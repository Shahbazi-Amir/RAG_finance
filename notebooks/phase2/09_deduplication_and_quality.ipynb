هدف این است که Branch، HEAD، وضعیت Git، Notebookهای قبلی، ورودی پذیرفته‌شده NB08 و فایل‌های قراردادی مرتبط مستقیماً از Repository شناسایی شوند.

<li>بررسی Branch فعلی.</li>
<li>ثبت HEAD فعلی.</li>
<li>بررسی <code>git status</code>.</li>
<li>بررسی وجود Notebookهای 07، 08 و 09.</li>
<li>بررسی وجود Dataset نرمال‌شده NB08.</li>

<h4>معیار PASS</h4>

<p>
Repository باید <code>RAG_finance</code> و Branch باید <code>rag-phase2</code> باشد.
Artifact ورودی اصلی NB08، Schema و Acceptance آن باید روی Disk قابل شناسایی باشند.
اگر هرکدام از ورودی‌های الزامی مفقود باشد، مرحله <code>FAIL</code> است و قبل از ادامه باید همان مورد بررسی شود.

```python
# Code Cell 01

from pathlib import Path
import subprocess

def run_git(*args):
    result = subprocess.run(
        ["git", *args],
        capture_output=True,
        text=True,
        check=True,
    )
    return result.stdout.strip()

repo_root = Path(run_git("rev-parse", "--show-toplevel")).resolve()
branch = run_git("branch", "--show-current")
head = run_git("rev-parse", "HEAD")
git_status = run_git("status", "--short")

required_paths = {
    "Notebook 07": repo_root / "notebooks/phase2/07_unified_schema_and_lineage.ipynb",
    "Notebook 08": repo_root / "notebooks/phase2/08_persian_normalization.ipynb",
    "Notebook 09": repo_root / "notebooks/phase2/09_deduplication_and_quality.ipynb",
    "Normalized dataset": repo_root / "data/phase2/processed/documents_normalized_v1.jsonl",
    "Normalization sidecar": repo_root / "data/phase2/processed/persian_normalization_records_v1.jsonl",
    "Normalization report": repo_root / "reports/phase2/persian_normalization_report_v1.json",
    "Normalization acceptance": repo_root / "reports/phase2/persian_normalization_acceptance_v1.json",
    "Document schema": repo_root / "schemas/phase2/document.schema.json",
}

print("Repository root:", repo_root)
print("Repository name:", repo_root.name)
print("Branch:", branch)
print("HEAD:", head)

print("\nGit status:")
print(git_status if git_status else "CLEAN")

print("\nPath discovery:")
for label, path in required_paths.items():
    print(f"{label}: {'FOUND' if path.exists() else 'MISSING'}")
    print(" ", path.relative_to(repo_root))

mandatory_labels = [
    "Notebook 07",
    "Notebook 08",
    "Normalized dataset",
    "Normalization sidecar",
    "Normalization report",
    "Normalization acceptance",
    "Document schema",
]

repository_ok = repo_root.name == "RAG_finance"
branch_ok = branch == "rag-phase2"
mandatory_paths_ok = all(required_paths[label].exists() for label in mandatory_labels)

gate_pass = repository_ok and branch_ok and mandatory_paths_ok

print("\nValidation:")
print("Repository check:", "PASS" if repository_ok else "FAIL")
print("Branch check:", "PASS" if branch_ok else "FAIL")
print("Mandatory input paths:", "PASS" if mandatory_paths_ok else "FAIL")
print("NB09 DISCOVERY START GATE:", "PASS" if gate_pass else "FAIL")
```

</p>

<p>
هم‌زمان، ورودی پذیرفته‌شده NB08 مستقیماً از Disk دوباره Hash می‌شود تا مطمئن شویم Notebook 09 دقیقاً روی همان Artifactهایی شروع می‌شود که در Acceptance مرحله قبل ثبت شده‌اند.
این مرحله همچنان فقط <strong>Discovery و Revalidation</strong> است و هیچ فایلی را تغییر نمی‌دهد.

<h4>بررسی‌های این مرحله</h4>

<ul>
<li>پیداکردن تمام فایل‌های دارای نام <code>09_deduplication_and_quality.ipynb</code> در Repository.</li>
<li>بررسی وضعیت Git و tracked/untracked بودن هر Candidate.</li>
<li>محاسبه SHA-256 و Size ورودی اصلی NB08.</li>
<li>محاسبه SHA-256 فایل Sidecar، Technical Report، Acceptance و Schema.</li>
<li>مقایسه Hashها با مقادیر پذیرفته‌شده پایان NB08.</li>
<li>خواندن وضعیت نهایی Acceptance و Scope boundary مستقیماً از فایل JSON.</li>
</ul>

<h4>معیار PASS</h4>

<ul>
<li>Hash تمام Artifactهای NB08 با Acceptance ثبت‌شده قبلی منطبق باشد.</li>
<li>Acceptance نهایی NB08 برابر <code>PASS</code> باشد.</li>
<li>Scope boundary نشان دهد NB08 کامل و NB09 هنوز شروع نشده است.</li>
<li>وضعیت واقعی فایل یا فایل‌های Notebook 09 بدون حدس مشخص شود.</li>
</ul>

<p>
اگر Hash یا Acceptance ناسازگار باشد، وارد Discovery الگوریتم‌ها و Logicهای Deduplication نمی‌شویم و ابتدا همان ناسازگاری بررسی می‌شود.
</p>

</div>

```python
# Code Cell 02

from pathlib import Path
import hashlib
import json
import subprocess

def sha256_file(path, chunk_size=1024 * 1024):
    digest = hashlib.sha256()
    with path.open("rb") as handle:
        while True:
            chunk = handle.read(chunk_size)
            if not chunk:
                break
            digest.update(chunk)
    return digest.hexdigest()

def git_output(*args):
    return subprocess.run(
        ["git", *args],
        cwd=repo_root,
        capture_output=True,
        text=True,
        check=True,
    ).stdout.strip()

expected = {
    "Normalized dataset": {
        "path": repo_root / "data/phase2/processed/documents_normalized_v1.jsonl",
        "sha256": "9d72c709f81b1218c5aac064052c7ed1da34869f671e6e32762864e0ad70ea40",
        "size": 63804043,
    },
    "Normalization sidecar": {
        "path": repo_root / "data/phase2/processed/persian_normalization_records_v1.jsonl",
        "sha256": "ea73c8e9b1547c11c58a0f5b845a0ca8e5932a6d72392ff81feb44f9404e6ef7",
        "size": 25084621,
    },
    "Normalization report": {
        "path": repo_root / "reports/phase2/persian_normalization_report_v1.json",
        "sha256": "bd3ffd58bb97dd460970aa9bd9043bf85690214d4614557dc384de611dd761ca",
        "size": 7402,
    },
    "Normalization acceptance": {
        "path": repo_root / "reports/phase2/persian_normalization_acceptance_v1.json",
        "sha256": "e4de079bda42e34c1fd05ecbb3a7d2a4101fe0ccfe4e2a3560dd60fe354d57bc",
        "size": 3637,
    },
    "Document schema": {
        "path": repo_root / "schemas/phase2/document.schema.json",
        "sha256": "3f577c71e8faa3b50c231bd4163cb5425158d0390e5fe5fcf39e00f530baa755",
        "size": None,
    },
}

print("NB09 notebook candidates:")
nb09_candidates = sorted(repo_root.rglob("09_deduplication_and_quality.ipynb"))

for path in nb09_candidates:
    relative = path.relative_to(repo_root)
    tracked = bool(git_output("ls-files", "--", str(relative)))
    status = git_output("status", "--short", "--", str(relative))
    print(f"- {relative}")
    print("  Tracked:", "YES" if tracked else "NO")
    print("  Git status:", status if status else "CLEAN / NO ENTRY")
    print("  Size:", path.stat().st_size, "bytes")
    print("  SHA-256:", sha256_file(path))

print("\nNB08 artifact revalidation:")

artifact_results = {}

for label, info in expected.items():
    path = info["path"]
    actual_hash = sha256_file(path)
    actual_size = path.stat().st_size

    hash_ok = actual_hash == info["sha256"]
    size_ok = info["size"] is None or actual_size == info["size"]

    artifact_results[label] = hash_ok and size_ok

    print(f"\n{label}:")
    print(" Path:", path.relative_to(repo_root))
    print(" Size:", actual_size)
    print(" SHA-256:", actual_hash)
    print(" Hash check:", "PASS" if hash_ok else "FAIL")
    print(" Size check:", "PASS" if size_ok else "FAIL")

acceptance_path = expected["Normalization acceptance"]["path"]

with acceptance_path.open("r", encoding="utf-8") as handle:
    acceptance = json.load(handle)

print("\nAcceptance discovery:")

for key in [
    "acceptance",
    "validation_status",
    "status",
    "scope_boundary",
    "policy_version",
    "policy_revision",
]:
    if key in acceptance:
        print(f"{key}: {acceptance[key]}")

print("\nTop-level acceptance keys:")
print(sorted(acceptance.keys()))

all_artifacts_ok = all(artifact_results.values())

print("\nArtifact integrity gate:", "PASS" if all_artifacts_ok else "FAIL")
print("NB09 CANDIDATE COUNT:", len(nb09_candidates))
print(
    "CODE CELL 02 GATE:",
    "PASS" if all_artifacts_ok and len(nb09_candidates) >= 1 else "FAIL",
)
```

و هیچ وضعیت ثبت‌شده‌ای وجود ندارد که شروع Discovery در NB09 را ممنوع کند.

اگر ساختار یا مقدارهای واقعی با Handoff قبلی ناسازگار باشند، مرحله <code>FAIL</code> خواهد بود و قبل از ادامه همان اختلاف بررسی می‌شود.

```python
# Code Cell 03

import json

acceptance_path = repo_root / "reports/phase2/persian_normalization_acceptance_v1.json"
report_path = repo_root / "reports/phase2/persian_normalization_report_v1.json"

with acceptance_path.open("r", encoding="utf-8") as handle:
    acceptance = json.load(handle)

with report_path.open("r", encoding="utf-8") as handle:
    report = json.load(handle)

print("=== NB08 ACCEPTANCE ===")
print("Acceptance status:", acceptance.get("acceptance_status"))
print("Acceptance type:", acceptance.get("acceptance_type"))
print("Acceptance version:", acceptance.get("acceptance_version"))

decision = acceptance.get("decision", {})
processing_boundaries = acceptance.get("processing_boundaries", {})

print("\nDecision:")
print(json.dumps(decision, ensure_ascii=False, indent=2))

print("\nProcessing boundaries:")
print(json.dumps(processing_boundaries, ensure_ascii=False, indent=2))

print("\nPolicy:")
print(json.dumps(acceptance.get("policy"), ensure_ascii=False, indent=2))

print("\nPopulation:")
print(json.dumps(acceptance.get("population"), ensure_ascii=False, indent=2))

print("\nQuality gates:")
print(json.dumps(acceptance.get("quality_gates"), ensure_ascii=False, indent=2))

print("\n=== NB08 TECHNICAL REPORT ===")

print("Top-level report keys:")
print(sorted(report.keys()))

for key in [
    "validation_status",
    "status",
    "policy",
    "population",
    "processing_boundaries",
    "decision",
    "summary",
]:
    if key in report:
        print(f"\n{key}:")
        print(json.dumps(report[key], ensure_ascii=False, indent=2))

acceptance_pass = acceptance.get("acceptance_status") == "PASS"
decision_pass = decision.get("status") == "PASS"
next_notebook_allowed = decision.get("next_notebook_allowed") is True
scope_boundary = decision.get("scope_boundary")

scope_boundary_ok = scope_boundary == "NB08_COMPLETE_NB09_NOT_STARTED"

dedup_not_performed = (
    processing_boundaries.get("deduplication") == "NOT_PERFORMED_IN_NB08"
)

print("\nValidation:")
print("NB08 acceptance:", "PASS" if acceptance_pass else "FAIL")
print("NB08 decision:", "PASS" if decision_pass else "FAIL")
print(
    "Next notebook allowed:",
    "PASS" if next_notebook_allowed else "FAIL",
)
print(
    "NB08 → NB09 scope boundary:",
    "PASS" if scope_boundary_ok else "FAIL",
)
print(
    "Deduplication untouched in NB08:",
    "PASS" if dedup_not_performed else "FAIL",
)

cell_gate = all(
    [
        acceptance_pass,
        decision_pass,
        next_notebook_allowed,
        scope_boundary_ok,
        dedup_not_performed,
    ]
)

print("CODE CELL 03 GATE:", "PASS" if cell_gate else "FAIL")
```

</p>

<p>
این مرحله فقط وقتی FAIL محسوب می‌شود که Repository قابل پیمایش نباشد یا Discovery به دلیل خطای اجرایی کامل نشود.
</p>

</div>

```python
# Code Cell 04

from pathlib import Path
import re

def relative_files(base):
    if not base.exists():
        return []
    return sorted(
        path.relative_to(repo_root)
        for path in base.rglob("*")
        if path.is_file()
    )

src_phase2 = repo_root / "src/phase2"
tests_root = repo_root / "tests"
processed_root = repo_root / "data/phase2/processed"
reports_root = repo_root / "reports/phase2"
notebooks_root = repo_root / "notebooks"

print("=== STRUCTURE DISCOVERY ===")

for label, base in [
    ("src/phase2", src_phase2),
    ("tests", tests_root),
    ("data/phase2/processed", processed_root),
    ("reports/phase2", reports_root),
]:
    files = relative_files(base)

    print(f"\n{label}:")
    print("Exists:", "YES" if base.exists() else "NO")
    print("Files:", len(files))

    for path in files[:120]:
        print("-", path)

    if len(files) > 120:
        print(f"... {len(files) - 120} more files not displayed")

search_terms = [
    "dedup",
    "duplicate",
    "near_duplicate",
    "near-duplicate",
    "similarity",
    "quarantine",
    "quality",
]

search_extensions = {
    ".py",
    ".ipynb",
    ".md",
    ".json",
    ".jsonl",
    ".yaml",
    ".yml",
    ".toml",
}

excluded_parts = {
    ".git",
    ".venv",
    "venv",
    "__pycache__",
    ".ipynb_checkpoints",
}

candidate_files = []

for path in repo_root.rglob("*"):
    if not path.is_file():
        continue

    if path.suffix.lower() not in search_extensions:
        continue

    if any(part in excluded_parts for part in path.parts):
        continue

    name_lower = path.name.lower()

    name_hits = [
        term for term in search_terms
        if term in name_lower
    ]

    content_hits = []

    if path.stat().st_size <= 5_000_000:
        try:
            text = path.read_text(encoding="utf-8", errors="ignore").lower()
            content_hits = [
                term for term in search_terms
                if re.search(re.escape(term), text)
            ]
        except OSError:
            content_hits = []

    hits = sorted(set(name_hits + content_hits))

    if hits:
        candidate_files.append(
            (
                path.relative_to(repo_root),
                hits,
                path.stat().st_size,
            )
        )

candidate_files.sort(key=lambda item: str(item[0]))

print("\n=== DEDUP / QUALITY CANDIDATES ===")
print("Candidate files:", len(candidate_files))

for path, hits, size in candidate_files:
    print(f"- {path}")
    print("  Matches:", ", ".join(hits))
    print("  Size:", size, "bytes")

print("\n=== CONVENTION SNAPSHOT ===")

processed_files = relative_files(processed_root)
report_files = relative_files(reports_root)

print("\nProcessed artifacts:")
for path in processed_files:
    print("-", path)

print("\nPhase 2 reports:")
for path in report_files:
    print("-", path)

print("\nValidation:")
print("Repository traversal: PASS")
print("src/phase2 discovered:", "YES" if src_phase2.exists() else "NO")
print("tests discovered:", "YES" if tests_root.exists() else "NO")
print("Dedup/quality candidates:", len(candidate_files))
print("CODE CELL 04 GATE: PASS")
```

<h3>Discovery قرارداد داده و تصمیم‌های Deduplication مراحل قبلی</h3>

<p>
بررسی Repository نشان داد در وضعیت فعلی
<code>src/phase2/</code>
و
<code>tests/</code>
وجود ندارند و منطق مرتبط با Deduplication و Quality عمدتاً در Notebookها، Manifestها، Artifactهای Interim و Acceptance Reportهای مراحل قبلی ثبت شده است.
</p>

<p>
پیش از طراحی Deduplication سراسری NB09 باید مشخص شود هر Common Document واقعاً چه فیلدهایی دارد و مراحل Website، Book و Transcript قبلاً چه تصمیم‌هایی درباره Duplicate، Canonical record، Quarantine یا Quality گرفته‌اند.
</p>

<p>
این مرحله فقط <strong>Discovery</strong> انجام می‌دهد. هیچ تصمیم Dedup جدیدی اعمال نمی‌شود و هیچ Metadata قبلی تغییر نمی‌کند.

<h4>موارد مورد بررسی</h4>

<ul>
<li>فیلدهای واقعی Common Document Schema.</li>
<li>ساختار واقعی یک Document از هر سه Source در Dataset نرمال‌شده.</li>

<li>کلیدهای مرتبط با Duplicate، Dedup، Canonical، Quarantine، Quality و Similarity در Manifestهای Book و Transcript.</li>
<li>کلیدهای مرتبط در Artifactهای Book ingestion.</li>
<li>تصمیم‌های ثبت‌شده در Acceptance Reportهای Website، Book و Transcript.</li>
</ul>

<h4>دلیل این بررسی</h4>

<p>
NB09 نباید تصمیم‌های Canonical یا Provenance مراحل قبلی را دوباره اختراع کند.
اگر مثلاً یک Edition کتاب یا یک Transcript پیش‌تر به‌عنوان Canonical یا Quarantined تعیین شده باشد،
Deduplication سراسری باید آن تصمیم را به‌عنوان Evidence ورودی بشناسد و فقط در صورت وجود دلیل جدید آن را بررسی کند.
</p>

<h4>معیار PASS</h4>

<p>
ساختار واقعی Common Document و Metadataهای مرتبط با تصمیم‌های قبلی برای هر سه Source باید بدون حدس قابل مشاهده باشد.
نبود یک فیلد خاص نیز نتیجه معتبر Discovery است.
</p>

<p>
این مرحله فقط در صورت خطای خواندن Artifactها یا ناتوانی در استخراج ساختار داده <code>FAIL</code> محسوب می‌شود.
</p>

```python
# Code Cell 05

import json
from collections import defaultdict

schema_path = repo_root / "schemas/phase2/document.schema.json"
normalized_path = repo_root / "data/phase2/processed/documents_normalized_v1.jsonl"

book_manifest_path = repo_root / "data/phase2/manifests/book_manifest_v1.json"
transcript_manifest_path = repo_root / "data/phase2/manifests/transcript_manifest_v1.json"
book_ingested_v2_path = repo_root / "data/phase2/interim/book_ingested_v2.json"

acceptance_paths = {
    "website": repo_root / "reports/phase2/website_acceptance_v1.json",
    "book_inventory": repo_root / "reports/phase2/book_inventory_acceptance_v1.json",
    "book_ingestion": repo_root / "reports/phase2/book_ingestion_acceptance_v1.json",
    "transcript_inventory": repo_root / "reports/phase2/transcript_inventory_acceptance_v1.json",
    "transcript_ingestion": repo_root / "reports/phase2/transcript_ingestion_acceptance_v1.json",
    "unified_schema": repo_root / "reports/phase2/unified_schema_lineage_acceptance_v1.json",
}

search_terms = (
    "dedup",
    "duplicate",
    "canonical",
    "quarantine",
    "quality",
    "similarity",
    "version",
    "edition",
)

def load_json(path):
    with path.open("r", encoding="utf-8") as handle:
        return json.load(handle)

def collect_matching_paths(obj, terms, path="$", results=None):
    if results is None:
        results = []

    if isinstance(obj, dict):
        for key, value in obj.items():
            current = f"{path}.{key}"
            key_lower = str(key).lower()

            if any(term in key_lower for term in terms):
                results.append((current, value))

            collect_matching_paths(value, terms, current, results)

    elif isinstance(obj, list):
        for index, value in enumerate(obj[:50]):
            collect_matching_paths(
                value,
                terms,
                f"{path}[{index}]",
                results,
            )

    return results

def compact_value(value, limit=400):
    rendered = json.dumps(value, ensure_ascii=False)
    if len(rendered) > limit:
        return rendered[:limit] + "... <truncated>"
    return rendered

schema = load_json(schema_path)

print("=== COMMON DOCUMENT SCHEMA ===")
print("Schema title:", schema.get("title"))
print("Required fields:")
print(schema.get("required"))

properties = schema.get("properties", {})
print("\nTop-level properties:")
for key in sorted(properties):
    field_type = properties[key].get("type")
    print(f"- {key}: {field_type}")

samples = {}

with normalized_path.open("r", encoding="utf-8") as handle:
    for line in handle:
        record = json.loads(line)
        source_type = record.get("source_type")

        if source_type in {"website", "book", "transcript"} and source_type not in samples:
            samples[source_type] = record

        if len(samples) == 3:
            break

print("\n=== NORMALIZED COMMON DOCUMENT SAMPLES ===")

for source_type in ["website", "book", "transcript"]:
    record = samples.get(source_type)

    print(f"\nSOURCE: {source_type}")

    if record is None:
        print("Sample: NOT FOUND")
        continue

    print("Top-level keys:")
    print(sorted(record.keys()))

    for field in [
        "document_id",
        "source_id",
        "source_type",
        "text_sha256",
        "raw_content_hash",
    ]:
        print(f"{field}:", compact_value(record.get(field)))

    for field in [
        "lineage",
        "attribution",
        "rights",
        "source_metadata",
    ]:
        value = record.get(field)
        print(f"\n{field}:")
        if isinstance(value, dict):
            print("Keys:", sorted(value.keys()))
            print(compact_value(value, 1200))
        else:
            print(compact_value(value, 1200))

artifact_paths = {
    "book_manifest_v1": book_manifest_path,
    "transcript_manifest_v1": transcript_manifest_path,
    "book_ingested_v2": book_ingested_v2_path,
}

print("\n=== PRIOR DEDUP / QUALITY METADATA ===")

for label, path in artifact_paths.items():
    print(f"\n{label}:")
    print("Path:", path.relative_to(repo_root))
    print("Exists:", "YES" if path.exists() else "NO")

    if not path.exists():
        continue

    data = load_json(path)
    matches = collect_matching_paths(data, search_terms)

    unique_matches = []
    seen = set()

    for key_path, value in matches:
        normalized_key_path = key_path
        if normalized_key_path not in seen:
            seen.add(normalized_key_path)
            unique_matches.append((key_path, value))

    print("Matched key paths:", len(unique_matches))

    for key_path, value in unique_matches[:80]:
        print("-", key_path)
        print(" ", compact_value(value))

    if len(unique_matches) > 80:
        print(f"... {len(unique_matches) - 80} additional matches not displayed")

print("\n=== ACCEPTANCE DECISIONS ===")

for label, path in acceptance_paths.items():
    print(f"\n{label}:")
    print("Path:", path.relative_to(repo_root))

    if not path.exists():
        print("Exists: NO")
        continue

    data = load_json(path)

    print("Exists: YES")
    print("Top-level keys:", sorted(data.keys()))

    matches = collect_matching_paths(data, search_terms)

    print("Relevant key paths:", len(matches))

    for key_path, value in matches[:60]:
        print("-", key_path)
        print(" ", compact_value(value))

    if len(matches) > 60:
        print(f"... {len(matches) - 60} additional matches not displayed")

source_samples_ok = all(
    source_type in samples
    for source_type in ["website", "book", "transcript"]
)

print("\nValidation:")
print(
    "Common Document schema loaded:",
    "PASS" if bool(properties) else "FAIL",
)
print(
    "Website/Book/Transcript samples found:",
    "PASS" if source_samples_ok else "FAIL",
)
print(
    "Prior artifact discovery completed:",
    "PASS",
)

cell_gate = bool(properties) and source_samples_ok

print("CODE CELL 05 GATE:", "PASS" if cell_gate else "FAIL")
```

<h3>پروفایل Exact Duplicate در Dataset نرمال‌شده</h3>

<p>
Discovery مراحل قبلی نشان داد که Deduplication سراسری هنوز انجام نشده است.
اکنون اولین بررسی محتوایی NB09 روی Dataset نرمال‌شده انجام می‌شود تا مشخص شود چند Document دقیقاً متن یکسان دارند.
</p>

<p>
برای این بررسی از
<code>text_sha256</code>
استفاده می‌شود، زیرا این Hash مستقیماً از متن نرمال‌شده هر Common Document ساخته شده است.
در مقابل،
<code>raw_content_hash</code>
لزوماً Hash واحد Document نیست؛ برای نمونه چند Paragraph کتاب یا چند Utterance یک Transcript می‌توانند عمداً به یک فایل Raw مشترک متصل باشند.
بنابراین تکرار
<code>raw_content_hash</code>
در این مرحله به‌عنوان Duplicate محتوایی تلقی نمی‌شود.
</p>

<p>
این مرحله فقط گروه‌های Exact Duplicate را شناسایی و طبقه‌بندی می‌کند.
هیچ Document حذف، Merge، Canonicalize یا Quarantine نمی‌شود.
مالکیت، Attribution، Rights و Lineage نیز بدون تغییر باقی می‌مانند.
</p>

<h4>طبقه‌بندی گروه‌های تکراری</h4>

<ul>
<li><strong>Same Source ID:</strong> متن یکسان در چند Document متعلق به همان Source والد.</li>
<li><strong>Same Source Type / Different Source IDs:</strong> متن یکسان در چند Source مستقل از یک خانواده، مانند دو Book Edition یا دو Episode.</li>
<li><strong>Cross-source:</strong> متن دقیقاً یکسان میان Website، Book یا Transcript.</li>
</ul>

<h4>خروجی مورد انتظار</h4>

<ul>
<li>تعداد کل Documentها و Hashهای یکتا.</li>
<li>تعداد گروه‌ها و اعضای Exact Duplicate.</li>
<li>توزیع Duplicateها به تفکیک Source Type.</li>
<li>تعداد گروه‌های Same Source ID، Same Source Type و Cross-source.</li>
<li>نمونه‌های محدود و قابل بررسی انسانی از گروه‌های تکراری.</li>
</ul>

<h4>معیار PASS</h4>

<p>
تمام رکوردها باید
<code>text_sha256</code>
معتبر داشته باشند و شمارش گروه‌های Exact Duplicate به‌صورت deterministic کامل شود.
وجود Duplicate به معنی FAIL نیست؛ Duplicateها در این مرحله Evidence برای تصمیم‌های بعدی هستند.
</p>

</div>

```python
# Code Cell 06

from pathlib import Path
from collections import Counter, defaultdict
import hashlib
import json
import subprocess

def find_repo_root():
    result = subprocess.run(
        ["git", "rev-parse", "--show-toplevel"],
        capture_output=True,
        text=True,
        check=True,
    )
    return Path(result.stdout.strip()).resolve()

repo_root = find_repo_root()
input_path = repo_root / "data/phase2/processed/documents_normalized_v1.jsonl"

records = []

with input_path.open("r", encoding="utf-8") as handle:
    for line_number, line in enumerate(handle, start=1):
        record = json.loads(line)

        records.append(
            {
                "line_number": line_number,
                "document_id": record.get("document_id"),
                "source_type": record.get("source_type"),
                "source_id": record.get("source_id"),
                "text_sha256": record.get("text_sha256"),
                "raw_content_hash": record.get("raw_content_hash"),
                "title": record.get("title"),
                "text": record.get("text", ""),
            }
        )

invalid_text_hashes = [
    row
    for row in records
    if not isinstance(row["text_sha256"], str)
    or len(row["text_sha256"]) != 64
]

groups = defaultdict(list)

for row in records:
    groups[row["text_sha256"]].append(row)

duplicate_groups = {
    text_hash: members
    for text_hash, members in groups.items()
    if len(members) > 1
}

duplicate_members = sum(
    len(members)
    for members in duplicate_groups.values()
)

duplicate_members_by_source = Counter()
group_classification = Counter()

classified_groups = []

for text_hash, members in duplicate_groups.items():
    source_types = {member["source_type"] for member in members}
    source_ids = {member["source_id"] for member in members}

    for member in members:
        duplicate_members_by_source[member["source_type"]] += 1

    if len(source_types) > 1:
        relation = "CROSS_SOURCE_TYPE"
    elif len(source_ids) > 1:
        relation = "SAME_SOURCE_TYPE_DIFFERENT_SOURCE_ID"
    else:
        relation = "SAME_SOURCE_ID"

    group_classification[relation] += 1

    classified_groups.append(
        {
            "text_sha256": text_hash,
            "relation": relation,
            "member_count": len(members),
            "members": members,
        }
    )

classified_groups.sort(
    key=lambda item: (
        -item["member_count"],
        item["relation"],
        item["text_sha256"],
    )
)

raw_hash_groups = defaultdict(list)

for row in records:
    raw_hash_groups[row["raw_content_hash"]].append(row)

shared_raw_hash_groups = {
    raw_hash: members
    for raw_hash, members in raw_hash_groups.items()
    if len(members) > 1
}

print("=== EXACT DUPLICATE PROFILE ===")
print("Input:", input_path.relative_to(repo_root))
print("Documents:", len(records))
print("Unique text SHA-256:", len(groups))
print("Exact duplicate groups:", len(duplicate_groups))
print("Exact duplicate members:", duplicate_members)
print(
    "Documents outside exact-duplicate groups:",
    len(records) - duplicate_members,
)

print("\nDuplicate members by source type:")
for source_type in ["website", "book", "transcript"]:
    print(
        f"- {source_type}:",
        duplicate_members_by_source.get(source_type, 0),
    )

print("\nDuplicate group classification:")
for relation in [
    "SAME_SOURCE_ID",
    "SAME_SOURCE_TYPE_DIFFERENT_SOURCE_ID",
    "CROSS_SOURCE_TYPE",
]:
    print(
        f"- {relation}:",
        group_classification.get(relation, 0),
    )

print("\nRaw-content hash context:")
print("Unique raw_content_hash values:", len(raw_hash_groups))
print(
    "Shared raw_content_hash groups:",
    len(shared_raw_hash_groups),
)
print(
    "Note: shared raw_content_hash is lineage evidence and is not treated as text duplication."
)

print("\n=== REPRESENTATIVE EXACT DUPLICATE GROUPS ===")

for index, group in enumerate(classified_groups[:20], start=1):
    print(f"\nGroup {index}")
    print("Relation:", group["relation"])
    print("Members:", group["member_count"])
    print("text_sha256:", group["text_sha256"])

    for member in group["members"][:12]:
        print(
            "-",
            member["source_type"],
            "|",
            member["source_id"],
            "|",
            member["document_id"],
        )

    preview = group["members"][0]["text"].replace("\n", " ").strip()
    print("Text preview:", preview[:240])

    if len(group["members"]) > 12:
        print(
            f"... {len(group['members']) - 12} additional members not displayed"
        )

print("\nValidation:")
print(
    "Valid text_sha256 for all records:",
    "PASS" if not invalid_text_hashes else "FAIL",
)
print(
    "Record count preserved:",
    "PASS" if len(records) == 19456 else "FAIL",
)

membership_check = sum(
    1
    for members in groups.values()
    for _ in members
) == len(records)

print(
    "Hash-group membership complete:",
    "PASS" if membership_check else "FAIL",
)

cell_gate = (
    not invalid_text_hashes
    and len(records) == 19456
    and membership_check
)

print("CODE CELL 06 GATE:", "PASS" if cell_gate else "FAIL")
```

<h3>تحلیل ماهیت Exact Duplicateها و بررسی کامل گروه‌های Cross-source</h3>

<p>
مرحله قبل وجود <code>192</code> گروه Exact Duplicate را اثبات کرد، اما مشاهده نمونه‌ها نشان داد که یکسان‌بودن Hash به‌تنهایی برای تصمیم حذف یا Quarantine کافی نیست.
بخشی از این گروه‌ها از متن‌های بسیار کوتاه، علائم، شماره‌ها، عبارت‌های مکالمه‌ای عمومی یا عناصر ساختاری کتاب تشکیل شده‌اند.
</p>

<p>
در این مرحله ماهیت این Duplicateها دقیق‌تر بررسی می‌شود تا میان «تکرار محتوای معنی‌دار» و «برابری متن‌های کم‌اطلاعات یا ساختاری» تفاوت قائل شویم.
این تفکیک برای جلوگیری از حذف اشتباه Evidence واقعی ضروری است.
</p>

<p>
همچنین تمام گروه‌های
<code>CROSS_SOURCE_TYPE</code>
به‌صورت کامل نمایش داده می‌شوند، زیرا Exact match میان Website، Book و Transcript می‌تواند نشانه بازنشر، نقل مستقیم، اقتباس یا تکرار واقعی محتوا باشد و باید پیش از هر Policy انسانی بررسی شود.
</p>

<p>
این مرحله همچنان فقط <strong>Discovery و Profiling</strong> است.
هیچ رکوردی حذف، Merge، Canonicalize یا Quarantine نمی‌شود.
</p>

<h4>بررسی‌های این مرحله</h4>

<ul>
<li>توزیع طول متن اعضای Exact Duplicate.</li>
<li>تفکیک گروه‌های دارای متن بسیار کوتاه از گروه‌های دارای محتوای طولانی‌تر.</li>
<li>شمارش Duplicateها بر اساس بازه طول متن.</li>
<li>نمایش کامل تمام گروه‌های Cross-source.</li>
<li>نمایش <code>document_id</code>، <code>source_id</code>، عنوان، Locator و Attribution هر عضو Cross-source.</li>
<li>بررسی اینکه گروه‌های Cross-source واقعاً محتوای معنی‌دار هستند یا فقط عبارت‌های عمومی کوتاه.</li>
</ul>

<h4>اصل تصمیم‌گیری</h4>

<p>
در این مرحله هیچ Threshold نهایی برای Quality یا Deduplication تعریف نمی‌شود.
بازه‌های طول فقط برای <strong>Profiling</strong> استفاده می‌شوند تا Corpus واقعی را بشناسیم.
هر Threshold اجرایی بعدی باید براساس همین شواهد و Inspection انسانی انتخاب شود.
</p>

<h4>معیار PASS</h4>

<p>
تمام <code>192</code> گروه Exact Duplicate باید در Profiling حساب شوند و هر <code>9</code> گروه Cross-source باید بدون حذف نمونه یا حدس Metadata قابل مشاهده باشند.
وجود محتوای کم‌اطلاعات یا Cross-source Duplicate به معنی FAIL نیست.
</p>

</div>

```python
# Code Cell 07

from collections import Counter
import json

def text_profile(text):
    text = text if isinstance(text, str) else ""
    stripped = text.strip()

    return {
        "chars": len(stripped),
        "words": len(stripped.split()),
        "lines": len(stripped.splitlines()) if stripped else 0,
        "text": stripped,
    }

length_bands = [
    ("EMPTY", 0, 0),
    ("1_3_CHARS", 1, 3),
    ("4_10_CHARS", 4, 10),
    ("11_30_CHARS", 11, 30),
    ("31_100_CHARS", 31, 100),
    ("101_300_CHARS", 101, 300),
    ("301_PLUS_CHARS", 301, None),
]

def assign_length_band(char_count):
    for label, minimum, maximum in length_bands:
        if char_count < minimum:
            continue
        if maximum is None or char_count <= maximum:
            return label
    return "UNCLASSIFIED"

group_length_distribution = Counter()
member_length_distribution = Counter()

profiled_groups = []

for group in classified_groups:
    representative_text = group["members"][0]["text"]
    profile = text_profile(representative_text)
    band = assign_length_band(profile["chars"])

    group_length_distribution[band] += 1
    member_length_distribution[band] += group["member_count"]

    profiled_groups.append(
        {
            **group,
            "chars": profile["chars"],
            "words": profile["words"],
            "length_band": band,
            "representative_text": profile["text"],
        }
    )

print("=== EXACT DUPLICATE LENGTH PROFILE ===")

for label, _, _ in length_bands:
    print(
        f"{label}:",
        "groups =", group_length_distribution.get(label, 0),
        "| members =", member_length_distribution.get(label, 0),
    )

print("\nTotal profiled duplicate groups:", len(profiled_groups))
print(
    "Total profiled duplicate members:",
    sum(group["member_count"] for group in profiled_groups),
)

cross_source_groups = [
    group
    for group in profiled_groups
    if group["relation"] == "CROSS_SOURCE_TYPE"
]

print("\n=== ALL CROSS-SOURCE EXACT DUPLICATE GROUPS ===")
print("Cross-source groups:", len(cross_source_groups))

record_lookup = {
    row["document_id"]: row
    for row in records
}

normalized_records_lookup = {}

with input_path.open("r", encoding="utf-8") as handle:
    for line in handle:
        record = json.loads(line)
        document_id = record.get("document_id")

        if document_id in {
            member["document_id"]
            for group in cross_source_groups
            for member in group["members"]
        }:
            normalized_records_lookup[document_id] = record

for index, group in enumerate(
    sorted(
        cross_source_groups,
        key=lambda item: (-item["chars"], item["text_sha256"]),
    ),
    start=1,
):
    print("\n" + "=" * 80)
    print(f"CROSS-SOURCE GROUP {index}")
    print("text_sha256:", group["text_sha256"])
    print("Members:", group["member_count"])
    print("Characters:", group["chars"])
    print("Words:", group["words"])
    print("Length band:", group["length_band"])
    print("Text:")
    print(group["representative_text"][:1500])

    if len(group["representative_text"]) > 1500:
        print("... <text truncated for display>")

    print("\nMembers:")

    for member in group["members"]:
        full_record = normalized_records_lookup.get(member["document_id"], {})

        print("\n- document_id:", member["document_id"])
        print("  source_type:", member["source_type"])
        print("  source_id:", member["source_id"])
        print("  title:", full_record.get("title"))

        locator = full_record.get("locator")
        attribution = full_record.get("attribution")

        print(
            "  locator:",
            json.dumps(locator, ensure_ascii=False),
        )

        if isinstance(attribution, dict):
            attribution_view = {
                "author": attribution.get("author"),
                "credited_author_name": attribution.get("credited_author_name"),
                "editorial_supervisor_name": attribution.get("editorial_supervisor_name"),
                "speaker": attribution.get("speaker"),
                "knowledge_owner": attribution.get("knowledge_owner"),
                "content_authority_type": attribution.get("content_authority_type"),
                "attribution_status": attribution.get("attribution_status"),
            }
        else:
            attribution_view = attribution

        print(
            "  attribution:",
            json.dumps(attribution_view, ensure_ascii=False),
        )

print("\n=== LONGER EXACT DUPLICATE GROUPS ===")

meaningful_length_groups = [
    group
    for group in profiled_groups
    if group["chars"] >= 100
]

meaningful_length_groups.sort(
    key=lambda item: (
        -item["chars"],
        -item["member_count"],
        item["text_sha256"],
    )
)

print(
    "Exact duplicate groups with at least 100 characters:",
    len(meaningful_length_groups),
)

for index, group in enumerate(meaningful_length_groups[:30], start=1):
    print(f"\nGroup {index}")
    print("Relation:", group["relation"])
    print("Members:", group["member_count"])
    print("Characters:", group["chars"])
    print("Words:", group["words"])
    print("text_sha256:", group["text_sha256"])
    print(
        "Text preview:",
        group["representative_text"][:500].replace("\n", " "),
    )

profile_group_count_ok = len(profiled_groups) == len(duplicate_groups)
profile_member_count_ok = (
    sum(group["member_count"] for group in profiled_groups)
    == duplicate_members
)
cross_source_count_ok = len(cross_source_groups) == 9

print("\nValidation:")
print(
    "All duplicate groups profiled:",
    "PASS" if profile_group_count_ok else "FAIL",
)
print(
    "All duplicate members accounted for:",
    "PASS" if profile_member_count_ok else "FAIL",
)
print(
    "Cross-source groups fully selected:",
    "PASS" if cross_source_count_ok else "FAIL",
)

cell_gate = (
    profile_group_count_ok
    and profile_member_count_ok
    and cross_source_count_ok
)

print("CODE CELL 07 GATE:", "PASS" if cell_gate else "FAIL")
```

<h3>بازبینی دقیق Exact Duplicateهای بلند و دارای محتوای بالقوه معنی‌دار</h3>

<p>
Profiling مرحله قبل نشان داد بیشتر Exact Duplicateهای Corpus بسیار کوتاه‌اند و Exact matchهای Cross-source نیز همگی متن‌های کوتاه یا عنوان‌مانند هستند.
در مقابل، فقط تعداد محدودی گروه دارای متن بلندتر از <code>100</code> کاراکتر هستند.
</p>

<p>
این گروه‌های بلند اهمیت بیشتری دارند، زیرا ممکن است یکی از چند حالت زیر باشند:
Boilerplate تکرارشونده برنامه،
عبارت ساختاری کتاب،
بخش معرفی یا اختتامیه ثابت،
یا محتوای آموزشی واقعاً تکرارشده.
این حالت‌ها از نظر Deduplication یکسان نیستند.
</p>

<p>
در این مرحله تمام گروه‌های Exact Duplicate با حداقل <code>100</code> کاراکتر با Metadata کامل‌تر بررسی می‌شوند.
هدف، شناخت ماهیت واقعی آن‌ها پیش از تعریف Policy است؛ بنابراین هنوز هیچ Rule برای حذف، Retain، Canonicalization یا Quarantine اعمال نمی‌شود.
</p>

<h4>موارد مورد بررسی</h4>

<ul>
<li>تمام اعضای هر گروه Exact Duplicate بلند.</li>
<li><code>source_id</code> و <code>document_id</code> هر عضو.</li>
<li>عنوان Source.</li>
<li>Locator کامل.</li>
<li>اطلاعات ساختاری مهم مانند Program، Episode، Heading، Work و Edition.</li>
<li>Attribution و Speaker برای جلوگیری از ادغام اشتباه مالکیت یا گفتار.</li>
<li>تعداد Source IDهای مستقل در هر گروه.</li>
<li>برخی ویژگی‌های توصیفی متن مانند تعداد کلمات و تنوع واژگان؛ فقط برای Profiling و نه Thresholdگذاری.</li>
</ul>

<h4>اصل مهم</h4>

<p>
حتی اگر متن طولانی در چند Episode یا Document دقیقاً یکسان باشد، این مرحله آن‌ها را خودکار حذف نمی‌کند.
برای مثال مقدمه ثابت یک برنامه رادیویی با یک پاراگراف آموزشی تکرارشده ماهیت یکسانی ندارد و باید Policy متفاوتی داشته باشند.
</p>

<h4>معیار PASS</h4>

<p>
هر ۱۳ گروه بلند شناسایی‌شده در مرحله قبل باید با تمام اعضا و Metadata لازم قابل مشاهده باشند.
هیچ تصمیم Deduplication در این مرحله شرط PASS نیست.

```python
# Code Cell 08

import json
import re
from collections import Counter

long_exact_groups = [
    group
    for group in profiled_groups
    if group["chars"] >= 100
]

long_exact_groups.sort(
    key=lambda item: (
        -item["chars"],
        -item["member_count"],
        item["text_sha256"],
    )
)

target_document_ids = {
    member["document_id"]
    for group in long_exact_groups
    for member in group["members"]
}

full_records = {}

with input_path.open("r", encoding="utf-8") as handle:
    for line in handle:
        record = json.loads(line)
        document_id = record.get("document_id")

        if document_id in target_document_ids:
            full_records[document_id] = record

def lexical_profile(text):
    stripped = text.strip()

    words = re.findall(
        r"[\w\u0600-\u06FF\u200c]+",
        stripped,
        flags=re.UNICODE,
    )

    unique_words = set(words)

    return {
        "characters": len(stripped),
        "words": len(words),
        "unique_words": len(unique_words),
        "unique_word_ratio": (
            round(len(unique_words) / len(words), 4)
            if words
            else None
        ),
    }

def selected_source_metadata(record):
    metadata = record.get("source_metadata")

    if not isinstance(metadata, dict):
        return metadata

    keys = [
        "work_id",
        "edition_id",
        "paragraph_number",
        "is_heading",
        "heading_level",
        "heading_path",
        "book_unit_type",
        "program",
        "program_id",
        "utterance_index",
        "record_type",
        "speaker",
        "timestamps",
        "merge",
    ]

    return {
        key: metadata.get(key)
        for key in keys
        if key in metadata
    }

print("=== LONG EXACT DUPLICATE DEEP INSPECTION ===")
print("Groups:", len(long_exact_groups))
print("Documents involved:", len(target_document_ids))

for index, group in enumerate(long_exact_groups, start=1):
    text = group["representative_text"]
    lexical = lexical_profile(text)

    source_ids = sorted({
        member["source_id"]
        for member in group["members"]
    })

    source_types = sorted({
        member["source_type"]
        for member in group["members"]
    })

    print("\n" + "=" * 100)
    print(f"GROUP {index}")
    print("Relation:", group["relation"])
    print("text_sha256:", group["text_sha256"])
    print("Members:", group["member_count"])
    print("Source types:", source_types)
    print("Unique source IDs:", len(source_ids))

    for source_id in source_ids:
        print(" -", source_id)

    print("\nLexical profile:")
    for key, value in lexical.items():
        print(f"- {key}: {value}")

    print("\nFull text:")
    print(text)

    print("\nMember details:")

    for member_index, member in enumerate(group["members"], start=1):
        record = full_records.get(member["document_id"], {})

        attribution = record.get("attribution")

        if isinstance(attribution, dict):
            attribution_view = {
                "author": attribution.get("author"),
                "credited_author_name": attribution.get("credited_author_name"),
                "editorial_supervisor_name": attribution.get("editorial_supervisor_name"),
                "speaker": attribution.get("speaker"),
                "knowledge_owner": attribution.get("knowledge_owner"),
                "content_authority_type": attribution.get("content_authority_type"),
                "attribution_status": attribution.get("attribution_status"),
            }
        else:
            attribution_view = attribution

        print(f"\nMember {member_index}:")
        print(" document_id:", record.get("document_id"))
        print(" source_type:", record.get("source_type"))
        print(" source_id:", record.get("source_id"))
        print(" title:", record.get("title"))

        print(
            " locator:",
            json.dumps(
                record.get("locator"),
                ensure_ascii=False,
            ),
        )

        print(
            " source_metadata:",
            json.dumps(
                selected_source_metadata(record),
                ensure_ascii=False,
            ),
        )

        print(
            " attribution:",
            json.dumps(
                attribution_view,
                ensure_ascii=False,
            ),
        )

print("\nValidation:")

group_count_ok = len(long_exact_groups) == 13
document_coverage_ok = (
    len(full_records) == len(target_document_ids)
)

all_members_accounted = all(
    member["document_id"] in full_records
    for group in long_exact_groups
    for member in group["members"]
)

print(
    "Expected long exact groups:",
    "PASS" if group_count_ok else "FAIL",
)
print(
    "Full record coverage:",
    "PASS" if document_coverage_ok else "FAIL",
)
print(
    "All group members resolved:",
    "PASS" if all_members_accounted else "FAIL",
)

cell_gate = (
    group_count_ok
    and document_coverage_ok
    and all_members_accounted
)

print("CODE CELL 08 GATE:", "PASS" if cell_gate else "FAIL")
```

<h3>پروفایل Granularity و طول Documentها پیش از طراحی Near-duplicate</h3>

<p>
بررسی Exact Duplicateها نشان داد که واحد Document در سه Source ماهیت یکسانی ندارد:
Website معمولاً یک مقاله کامل است،
Book در سطح Paragraph نگهداری شده
و Transcript در سطح Segment یا Utterance قرار دارد.
</p>

<p>
بنابراین استفاده از یک روش یا Threshold یکسان برای Near-duplicate در هر سه Source می‌تواند نتیجه اشتباه تولید کند.
برای مثال شباهت زیاد میان دو مقاله کامل با شباهت میان دو عبارت کوتاه مکالمه‌ای معنای یکسانی ندارد.
</p>

<p>
پیش از انتخاب هر Similarity Metric، Candidate Generation روش، MinHash، N-gram، Edit Distance، Embedding یا Threshold باید Granularity و توزیع واقعی طول متن در هر Source اندازه‌گیری شود.
</p>

<p>
این مرحله فقط <strong>Corpus Profiling</strong> است و هیچ Near-duplicateای را اعلام نمی‌کند.
هیچ Algorithm یا Threshold اجرایی نیز در این مرحله انتخاب نمی‌شود.
</p>

<h4>موارد مورد بررسی</h4>

<ul>
<li>تعداد Documentهای هر Source.</li>
<li>توزیع طول متن به Character و Word.</li>
<li>Minimum، Maximum و Quantileهای طول هر Source.</li>
<li>تعداد متن‌های بسیار کوتاه و بلند در بازه‌های توصیفی.</li>
<li>انواع واحدهای Book از طریق <code>book_unit_type</code>.</li>
<li>انواع واحدهای Transcript از طریق <code>record_type</code>.</li>
<li>تعداد Work، Edition، Program و Source IDهای مستقل.</li>
<li>Coverage فیلدهای Locator که در تحلیل Near-duplicate بعدی برای حفظ Context لازم خواهند بود.</li>
</ul>

<h4>دلیل این بررسی</h4>

<p>
Near-duplicate باید به واحد محتوایی واقعی احترام بگذارد.
اگر Corpus نشان دهد Website، Book و Transcript از نظر طول و ساختار تفاوت بنیادی دارند،
مرحله بعد باید Candidate Generation و Similarity را برای هر Source جداگانه طراحی کند و سپس Cross-source را به‌عنوان مسئله مستقل بررسی کند.
</p>

<h4>معیار PASS</h4>

<p>
تمام <code>19,456</code> Document باید در Profiling حضور داشته باشند و برای هر سه Source آمار طول و ساختار قابل استخراج باشد.
وجود متن کوتاه یا Outlier به معنی FAIL نیست و فقط Evidence برای طراحی مرحله بعد محسوب می‌شود.
</p>

</div>

```python
# Code Cell 09

from pathlib import Path
from collections import Counter, defaultdict
import json
import math
import re
import subprocess

def find_repo_root():
    result = subprocess.run(
        ["git", "rev-parse", "--show-toplevel"],
        capture_output=True,
        text=True,
        check=True,
    )
    return Path(result.stdout.strip()).resolve()

def percentile(values, q):
    if not values:
        return None

    ordered = sorted(values)

    if len(ordered) == 1:
        return ordered[0]

    position = (len(ordered) - 1) * q
    lower = math.floor(position)
    upper = math.ceil(position)

    if lower == upper:
        return ordered[lower]

    fraction = position - lower

    return (
        ordered[lower] * (1 - fraction)
        + ordered[upper] * fraction
    )

def word_count(text):
    return len(
        re.findall(
            r"[\w\u0600-\u06FF\u200c]+",
            text,
            flags=re.UNICODE,
        )
    )

repo_root = find_repo_root()
input_path = repo_root / "data/phase2/processed/documents_normalized_v1.jsonl"

source_profiles = defaultdict(
    lambda: {
        "documents": 0,
        "char_lengths": [],
        "word_lengths": [],
        "source_ids": set(),
        "work_ids": set(),
        "edition_ids": set(),
        "program_ids": set(),
        "book_unit_types": Counter(),
        "record_types": Counter(),
        "locator_nonempty": 0,
    }
)

length_bands = [
    ("0_10", 0, 10),
    ("11_30", 11, 30),
    ("31_100", 31, 100),
    ("101_300", 101, 300),
    ("301_1000", 301, 1000),
    ("1001_3000", 1001, 3000),
    ("3001_PLUS", 3001, None),
]

length_band_counts = defaultdict(Counter)

total_records = 0

with input_path.open("r", encoding="utf-8") as handle:
    for line in handle:
        record = json.loads(line)
        total_records += 1

        source_type = record.get("source_type")
        text = record.get("text") or ""
        stripped = text.strip()

        chars = len(stripped)
        words = word_count(stripped)

        profile = source_profiles[source_type]

        profile["documents"] += 1
        profile["char_lengths"].append(chars)
        profile["word_lengths"].append(words)
        profile["source_ids"].add(record.get("source_id"))

        locator = record.get("locator")

        if isinstance(locator, dict) and locator:
            profile["locator_nonempty"] += 1

        metadata = record.get("source_metadata")

        if isinstance(metadata, dict):
            work_id = metadata.get("work_id")
            edition_id = metadata.get("edition_id")
            program_id = metadata.get("program_id")
            book_unit_type = metadata.get("book_unit_type")
            record_type = metadata.get("record_type")

            if work_id is not None:
                profile["work_ids"].add(work_id)

            if edition_id is not None:
                profile["edition_ids"].add(edition_id)

            if program_id is not None:
                profile["program_ids"].add(program_id)

            if book_unit_type is not None:
                profile["book_unit_types"][book_unit_type] += 1

            if record_type is not None:
                profile["record_types"][record_type] += 1

        for label, minimum, maximum in length_bands:
            if chars < minimum:
                continue

            if maximum is None or chars <= maximum:
                length_band_counts[source_type][label] += 1
                break

print("=== DOCUMENT GRANULARITY PROFILE ===")
print("Input:", input_path.relative_to(repo_root))
print("Total documents:", total_records)

for source_type in ["website", "book", "transcript"]:
    profile = source_profiles[source_type]

    chars = profile["char_lengths"]
    words = profile["word_lengths"]

    print("\n" + "=" * 80)
    print("SOURCE:", source_type)
    print("Documents:", profile["documents"])
    print("Unique source IDs:", len(profile["source_ids"]))
    print("Non-empty locator:", profile["locator_nonempty"])

    print("\nCharacter length:")
    print(" Min:", min(chars) if chars else None)
    print(" P25:", round(percentile(chars, 0.25), 2) if chars else None)
    print(" Median:", round(percentile(chars, 0.50), 2) if chars else None)
    print(" P75:", round(percentile(chars, 0.75), 2) if chars else None)
    print(" P90:", round(percentile(chars, 0.90), 2) if chars else None)
    print(" P95:", round(percentile(chars, 0.95), 2) if chars else None)
    print(" P99:", round(percentile(chars, 0.99), 2) if chars else None)
    print(" Max:", max(chars) if chars else None)

    print("\nWord length:")
    print(" Min:", min(words) if words else None)
    print(" P25:", round(percentile(words, 0.25), 2) if words else None)
    print(" Median:", round(percentile(words, 0.50), 2) if words else None)
    print(" P75:", round(percentile(words, 0.75), 2) if words else None)
    print(" P90:", round(percentile(words, 0.90), 2) if words else None)
    print(" P95:", round(percentile(words, 0.95), 2) if words else None)
    print(" P99:", round(percentile(words, 0.99), 2) if words else None)
    print(" Max:", max(words) if words else None)

    print("\nCharacter bands:")
    for label, _, _ in length_bands:
        print(
            f"- {label}:",
            length_band_counts[source_type].get(label, 0),
        )

    if profile["work_ids"]:
        print("\nBook structure:")
        print(" Works:", len(profile["work_ids"]))
        print(" Editions:", len(profile["edition_ids"]))
        print(
            " Unit types:",
            dict(sorted(profile["book_unit_types"].items())),
        )

    if profile["program_ids"]:
        print("\nTranscript structure:")
        print(" Programs:", len(profile["program_ids"]))
        print(
            " Record types:",
            dict(sorted(profile["record_types"].items())),
        )

print("\n=== SOURCE COMPARISON ===")

for source_type in ["website", "book", "transcript"]:
    profile = source_profiles[source_type]

    median_chars = percentile(
        profile["char_lengths"],
        0.50,
    )

    median_words = percentile(
        profile["word_lengths"],
        0.50,
    )

    print(
        source_type,
        "| documents =", profile["documents"],
        "| median chars =", round(median_chars, 2),
        "| median words =", round(median_words, 2),
        "| unique source IDs =", len(profile["source_ids"]),
    )

expected_distribution = {
    "website": 529,
    "book": 2701,
    "transcript": 16226,
}

distribution_ok = all(
    source_profiles[source_type]["documents"] == expected_count
    for source_type, expected_count in expected_distribution.items()
)

locator_coverage_ok = all(
    source_profiles[source_type]["locator_nonempty"]
    == source_profiles[source_type]["documents"]
    for source_type in expected_distribution
)

print("\nValidation:")
print(
    "Total record count:",
    "PASS" if total_records == 19456 else "FAIL",
)
print(
    "Source distribution preserved:",
    "PASS" if distribution_ok else "FAIL",
)
print(
    "Locator coverage:",
    "PASS" if locator_coverage_ok else "FAIL",
)

cell_gate = (
    total_records == 19456
    and distribution_ok
    and locator_coverage_ok
)

print("CODE CELL 09 GATE:", "PASS" if cell_gate else "FAIL")
```

<h3>Discovery قابلیت‌های محیط برای Benchmark روش‌های Near-duplicate</h3>

<p>
Profiling Corpus نشان داد که Granularity سه Source تفاوت بنیادی دارد:
میانه طول Website حدود <code>1511</code> کاراکتر،
Book حدود <code>31</code> کاراکتر
و Transcript حدود <code>44</code> کاراکتر است.
</p>

<p>
بنابراین پیش از انتخاب روش Near-duplicate باید مشخص شود محیط فعلی پروژه چه Libraryهای محاسباتی و Similarity را در اختیار دارد.
هدف این مرحله انتخاب Algorithm نیست؛ فقط قابلیت‌های واقعی Environment بررسی می‌شوند تا در مرحله R&amp;D بعدی بتوان چند روش قابل اجرا را به‌صورت مقایسه‌ای Benchmark کرد.
</p>

<p>
در این مرحله هیچ Package نصب یا Upgrade نمی‌شود و هیچ Dependency فایل پروژه تغییر نمی‌کند.
تنها Libraryهای موجود در همان Environment فعال Jupyter شناسایی می‌شوند.
</p>

<h4>دلیل این بررسی</h4>

<p>
Near-duplicate detection می‌تواند با روش‌های مختلفی مانند Token similarity، Character similarity، Edit similarity، Shingling یا Approximate candidate generation انجام شود.
اما انتخاب روش باید بر اساس Corpus واقعی، هزینه محاسباتی و کیفیت Candidateها باشد؛ نه صرفاً بر اساس حدس یا عادت.
</p>

<p>
به همین دلیل ابتدا مشخص می‌کنیم چه ابزارهایی واقعاً موجودند و سپس در مرحله بعد چند روش مناسب را روی نمونه‌های کنترل‌شده مقایسه خواهیم کرد.
</p>

<h4>معیار PASS</h4>

<p>
Cell باید Python Environment فعال را بدون تغییر بررسی کند و وضعیت هر Library را صریحاً به شکل
<code>AVAILABLE</code>
یا
<code>NOT AVAILABLE</code>
گزارش دهد.
نبود هیچ‌یک از Libraryهای اختیاری به معنی FAIL نیست.
</p>

```python
# Code Cell 10

import importlib
import importlib.metadata
import platform
import sys

packages = {
    "numpy": "numpy",
    "pandas": "pandas",
    "scikit-learn": "sklearn",
    "rapidfuzz": "rapidfuzz",
    "datasketch": "datasketch",
    "scipy": "scipy",
    "sentence-transformers": "sentence_transformers",
}

def package_version(distribution_name):
    try:
        return importlib.metadata.version(distribution_name)
    except importlib.metadata.PackageNotFoundError:
        return None

print("=== ACTIVE PYTHON ENVIRONMENT ===")
print("Python version:", sys.version.replace("\n", " "))
print("Python executable:", sys.executable)
print("Platform:", platform.platform())

print("\n=== NEAR-DUPLICATE TOOLING DISCOVERY ===")

availability = {}

for distribution_name, import_name in packages.items():
    try:
        module = importlib.import_module(import_name)
        available = True
        import_error = None
    except Exception as exc:
        module = None
        available = False
        import_error = f"{type(exc).__name__}: {exc}"

    version = package_version(distribution_name)

    availability[distribution_name] = {
        "available": available,
        "version": version,
        "import_error": import_error,
    }

    print(f"\n{distribution_name}:")
    print(" Status:", "AVAILABLE" if available else "NOT AVAILABLE")
    print(" Version:", version)

    if import_error is not None:
        print(" Import result:", import_error)

print("\n=== STANDARD-LIBRARY BASELINE ===")

stdlib_capabilities = {
    "difflib.SequenceMatcher": False,
    "collections": False,
    "hashlib": False,
    "re": False,
}

try:
    from difflib import SequenceMatcher
    stdlib_capabilities["difflib.SequenceMatcher"] = True
except Exception:
    pass

try:
    import collections
    stdlib_capabilities["collections"] = True
except Exception:
    pass

try:
    import hashlib
    stdlib_capabilities["hashlib"] = True
except Exception:
    pass

try:
    import re
    stdlib_capabilities["re"] = True
except Exception:
    pass

for name, available in stdlib_capabilities.items():
    print(
        f"- {name}:",
        "AVAILABLE" if available else "NOT AVAILABLE",
    )

stdlib_ok = all(stdlib_capabilities.values())

print("\nValidation:")
print(
    "Python runtime discovered:",
    "PASS" if bool(sys.executable) else "FAIL",
)
print(
    "Standard-library similarity baseline:",
    "PASS" if stdlib_ok else "FAIL",
)
print(
    "Optional packages inspected:",
    "PASS" if len(availability) == len(packages) else "FAIL",
)

cell_gate = (
    bool(sys.executable)
    and stdlib_ok
    and len(availability) == len(packages)
)

print("CODE CELL 10 GATE:", "PASS" if cell_gate else "FAIL")
```

هدف این Benchmark تعیین Threshold نهایی نیست؛ بلکه مقایسه حساسیت روش‌ها به انواع مختلف شباهت متنی است.

<h4>گروه‌های Benchmark</h4>

<ul>
<li><strong>EXACT:</strong> جفت‌هایی که <code>text_sha256</code> یکسان دارند.</li>
<li><strong>CONTROLLED_NEAR:</strong> نسخه‌های مصنوعی و فقط آزمایشی از متن واقعی که تغییرات محدود و قابل‌کنترل در آن‌ها ایجاد شده است.</li>
<li><strong>RANDOM_INDEPENDENT:</strong> جفت‌های تصادفی از Documentهای مستقل که فقط برای مشاهده Baseline توزیع Similarity استفاده می‌شوند و به‌عنوان Ground Truth قطعی منفی تلقی نمی‌شوند.</li>
</ul>

<p>
نسخه‌های <code>CONTROLLED_NEAR</code> هیچ‌گاه وارد Dataset یا Artifact نهایی نمی‌شوند.
آن‌ها فقط ابزار آزمایش روش هستند.
</p>

<h4>روش‌های مقایسه‌شونده</h4>

<ul>
<li><code>SequenceMatcher</code> روی متن.</li>
<li>Cosine Similarity مبتنی بر Word TF-IDF.</li>
<li>Cosine Similarity مبتنی بر Character TF-IDF.</li>
</ul>

<p>
پارامترهای این مرحله فقط برای Benchmark آزمایشی هستند و هنوز Policy تولید محسوب نمی‌شوند.
هیچ Threshold قبول یا رد Near-duplicate در این مرحله تعریف نخواهد شد.
</p>

<h4>هدف آزمایش</h4>

<p>
باید مشخص شود آیا روش‌های مختلف می‌توانند سه وضعیت
Exact،
تغییرات محدود،
و جفت‌های مستقل
را از یکدیگر تا حد قابل‌مشاهده تفکیک کنند یا خیر.
نتیجه این Benchmark مبنای طراحی Candidate Generation مرحله بعد خواهد بود.
</p>

<h4>معیار PASS</h4>

<p>
Benchmark باید به‌صورت deterministic ساخته شود،
تمام جفت‌ها دارای Label آزمایشی مشخص باشند،
هر سه روش برای تمام جفت‌ها Score تولید کنند
و هیچ NaN یا خطای محاسباتی وجود نداشته باشد.
بالا یا پایین بودن Score خاصی در این مرحله شرط PASS نیست.
</p>

```python
# Code Cell 11

from difflib import SequenceMatcher
from collections import Counter
import hashlib
import json
import random
import re

import numpy as np
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

RANDOM_SEED = 20260901
rng = random.Random(RANDOM_SEED)

# -------------------------------------------------------------------
# Load a compact representation of the normalized corpus independently
# -------------------------------------------------------------------

benchmark_records = []

with input_path.open("r", encoding="utf-8") as handle:
    for line in handle:
        record = json.loads(line)

        benchmark_records.append(
            {
                "document_id": record["document_id"],
                "source_type": record["source_type"],
                "source_id": record["source_id"],
                "text_sha256": record["text_sha256"],
                "text": record["text"].strip(),
            }
        )

# -------------------------------------------------------------------
# Exact-pair pool
# -------------------------------------------------------------------

hash_members = {}

for record in benchmark_records:
    hash_members.setdefault(
        record["text_sha256"],
        [],
    ).append(record)

exact_groups_for_benchmark = [
    members
    for members in hash_members.values()
    if len(members) >= 2
    and len(members[0]["text"]) >= 30
]

exact_pairs = []

for members in exact_groups_for_benchmark:
    first = members[0]
    second = members[1]

    exact_pairs.append(
        {
            "label": "EXACT",
            "left_id": first["document_id"],
            "right_id": second["document_id"],
            "left_text": first["text"],
            "right_text": second["text"],
        }
    )

exact_pairs.sort(
    key=lambda item: (
        item["left_id"],
        item["right_id"],
    )
)

exact_pairs = exact_pairs[:40]

# -------------------------------------------------------------------
# Controlled near-duplicate transformations
# These texts are NEVER written to project artifacts.
# -------------------------------------------------------------------

eligible_for_controlled = [
    record
    for record in benchmark_records
    if len(record["text"]) >= 300
]

eligible_for_controlled.sort(
    key=lambda item: item["document_id"]
)

controlled_source_records = rng.sample(
    eligible_for_controlled,
    min(40, len(eligible_for_controlled)),
)

def controlled_edit(text):
    """
    Deterministic, small lexical perturbation for benchmark use only.
    No output from this function becomes project data.
    """
    words = text.split()

    if len(words) < 20:
        return text

    # Remove approximately one word out of every 20 words.
    edited_words = [
        word
        for index, word in enumerate(words, start=1)
        if index % 20 != 0
    ]

    edited = " ".join(edited_words)

    # Remove one punctuation family to introduce a second limited change.
    edited = re.sub(r"[،؛]", " ", edited)
    edited = re.sub(r" {2,}", " ", edited).strip()

    return edited

controlled_pairs = []

for record in controlled_source_records:
    modified = controlled_edit(record["text"])

    if modified != record["text"]:
        controlled_pairs.append(
            {
                "label": "CONTROLLED_NEAR",
                "left_id": record["document_id"],
                "right_id": record["document_id"] + "::controlled_edit",
                "left_text": record["text"],
                "right_text": modified,
            }
        )

# -------------------------------------------------------------------
# Random independent baseline
# This is descriptive only, not authoritative negative ground truth.
# -------------------------------------------------------------------

independent_pairs = []

attempts = 0

while len(independent_pairs) < 40 and attempts < 5000:
    attempts += 1

    left, right = rng.sample(benchmark_records, 2)

    if left["text_sha256"] == right["text_sha256"]:
        continue

    if left["document_id"] == right["document_id"]:
        continue

    # Avoid tiny strings that would make the baseline uninformative.
    if len(left["text"]) < 30 or len(right["text"]) < 30:
        continue

    independent_pairs.append(
        {
            "label": "RANDOM_INDEPENDENT",
            "left_id": left["document_id"],
            "right_id": right["document_id"],
            "left_text": left["text"],
            "right_text": right["text"],
        }
    )

benchmark_pairs = (
    exact_pairs
    + controlled_pairs
    + independent_pairs
)

print("=== CONTROLLED BENCHMARK DATASET ===")
print("Random seed:", RANDOM_SEED)
print("Total benchmark pairs:", len(benchmark_pairs))

label_counts = Counter(
    pair["label"]
    for pair in benchmark_pairs
)

for label in [
    "EXACT",
    "CONTROLLED_NEAR",
    "RANDOM_INDEPENDENT",
]:
    print(
        f"- {label}:",
        label_counts.get(label, 0),
    )

# -------------------------------------------------------------------
# Similarity methods
# -------------------------------------------------------------------

all_texts = []

for pair in benchmark_pairs:
    all_texts.append(pair["left_text"])
    all_texts.append(pair["right_text"])

word_vectorizer = TfidfVectorizer(
    analyzer="word",
    ngram_range=(1, 2),
    lowercase=False,
    min_df=1,
)

char_vectorizer = TfidfVectorizer(
    analyzer="char_wb",
    ngram_range=(3, 5),
    lowercase=False,
    min_df=1,
)

word_matrix = word_vectorizer.fit_transform(all_texts)
char_matrix = char_vectorizer.fit_transform(all_texts)

benchmark_results = []

for index, pair in enumerate(benchmark_pairs):
    left_index = index * 2
    right_index = left_index + 1

    sequence_score = SequenceMatcher(
        None,
        pair["left_text"],
        pair["right_text"],
        autojunk=False,
    ).ratio()

    word_score = cosine_similarity(
        word_matrix[left_index],
        word_matrix[right_index],
    )[0, 0]

    char_score = cosine_similarity(
        char_matrix[left_index],
        char_matrix[right_index],
    )[0, 0]

    benchmark_results.append(
        {
            "label": pair["label"],
            "left_id": pair["left_id"],
            "right_id": pair["right_id"],
            "sequence_matcher": float(sequence_score),
            "word_tfidf_cosine": float(word_score),
            "char_tfidf_cosine": float(char_score),
        }
    )

# -------------------------------------------------------------------
# Descriptive summary
# -------------------------------------------------------------------

def score_summary(values):
    values = np.asarray(values, dtype=float)

    return {
        "min": float(np.min(values)),
        "p25": float(np.percentile(values, 25)),
        "median": float(np.median(values)),
        "p75": float(np.percentile(values, 75)),
        "max": float(np.max(values)),
    }

methods = [
    "sequence_matcher",
    "word_tfidf_cosine",
    "char_tfidf_cosine",
]

print("\n=== BENCHMARK SCORE DISTRIBUTIONS ===")

for label in [
    "EXACT",
    "CONTROLLED_NEAR",
    "RANDOM_INDEPENDENT",
]:
    subset = [
        result
        for result in benchmark_results
        if result["label"] == label
    ]

    print("\nLABEL:", label)
    print("Pairs:", len(subset))

    for method in methods:
        values = [
            result[method]
            for result in subset
        ]

        summary = score_summary(values)

        print(f"\n{method}:")
        for key, value in summary.items():
            print(
                f"  {key}:",
                round(value, 6),
            )

print("\n=== SAMPLE SCORES ===")

for result in benchmark_results[:15]:
    print(
        result["label"],
        "|",
        result["left_id"],
        "<->",
        result["right_id"],
    )
    print(
        " SequenceMatcher:",
        round(result["sequence_matcher"], 6),
        "| Word TF-IDF:",
        round(result["word_tfidf_cosine"], 6),
        "| Char TF-IDF:",
        round(result["char_tfidf_cosine"], 6),
    )

# -------------------------------------------------------------------
# Validation
# -------------------------------------------------------------------

all_scores_finite = all(
    np.isfinite(result[method])
    for result in benchmark_results
    for method in methods
)

labels_present = all(
    label_counts.get(label, 0) > 0
    for label in [
        "EXACT",
        "CONTROLLED_NEAR",
        "RANDOM_INDEPENDENT",
    ]
)

exact_integrity = all(
    pair["left_text"] == pair["right_text"]
    for pair in benchmark_pairs
    if pair["label"] == "EXACT"
)

controlled_integrity = all(
    pair["left_text"] != pair["right_text"]
    for pair in benchmark_pairs
    if pair["label"] == "CONTROLLED_NEAR"
)

print("\nValidation:")
print(
    "All benchmark labels present:",
    "PASS" if labels_present else "FAIL",
)
print(
    "Exact-pair integrity:",
    "PASS" if exact_integrity else "FAIL",
)
print(
    "Controlled-near integrity:",
    "PASS" if controlled_integrity else "FAIL",
)
print(
    "All similarity scores finite:",
    "PASS" if all_scores_finite else "FAIL",
)

cell_gate = (
    labels_present
    and exact_integrity
    and controlled_integrity
    and all_scores_finite
)

print(
    "CODE CELL 11 GATE:",
    "PASS" if cell_gate else "FAIL",
)
```

<div style="direction: rtl; white-space: normal; line-height: 1.7;">

<h3>اعتبارسنجی Failure Modeهای روش‌های Similarity پیش از انتخاب Candidate Generator</h3>

<p>
Benchmark کنترل‌شده نشان داد هر سه روش توانستند Controlled Near-duplicateها را از Random Independent pairها تا حد زیادی جدا کنند.
با این حال یک Failure Mode مهم در
<code>Word TF-IDF</code>
مشاهده شد:
برای بعضی Exact Duplicateها Score برابر <code>0</code> تولید شده است.
</p>

<p>
این رفتار باید قبل از انتخاب روش اصلی Near-duplicate بررسی شود.
احتمال اولیه این است که برخی متن‌ها فقط شامل علامت، نقطه یا واحدهایی باشند که Tokenizer پیش‌فرض Word TF-IDF برای آن‌ها هیچ Token معتبری تولید نمی‌کند؛ اما این فرض باید با داده واقعی تأیید شود.
</p>

<p>
در این مرحله سه روش از نظر Failure Mode بررسی می‌شوند.
هدف هنوز انتخاب Threshold یا حذف یک روش نیست؛ بلکه مشخص‌کردن محدودیت‌های واقعی هر روش است.
</p>

<h4>بررسی‌های این مرحله</h4>

<ul>
<li>شناسایی تمام Exact pairهایی که هر روش برای آن‌ها Score کمتر از <code>1</code> داده است.</li>
<li>بررسی متن واقعی این Failureها.</li>
<li>بررسی تعداد Tokenهای Word TF-IDF برای همان متن‌ها.</li>
<li>مقایسه پایداری سه روش روی Exact pairها.</li>
<li>بررسی Minimum margin میان Controlled Near و Random Independent در Benchmark فعلی.</li>
<li>بررسی اینکه آیا Scoreها در محدوده معتبر <code>0..1</code> قرار دارند.</li>
</ul>

<h4>اصل تصمیم‌گیری</h4>

<p>
یک Similarity method فقط به این دلیل که روی اکثر نمونه‌ها خوب عمل کرده پذیرفته نمی‌شود.
روش انتخابی باید Failure Mode شناخته‌شده داشته باشد و مشخص باشد در چه نوع Documentهایی قابل اعتماد نیست.
</p>

<p>
اگر Character TF-IDF یا SequenceMatcher در Exact integrity پایدارتر باشند، این موضوع فقط Evidence برای طراحی Candidate Generation مرحله بعد خواهد بود و هنوز به معنی تثبیت Threshold تولید نیست.
</p>

<h4>معیار PASS</h4>

<p>
تمام Failureهای Exact benchmark باید قابل توضیح و قابل ردیابی باشند،
تمام Scoreها باید در محدوده معتبر قرار داشته باشند
و خلاصه پایداری هر سه روش بدون NaN یا خطای محاسباتی تولید شود.
</p>

</div>

```python
# Code Cell 12

from collections import Counter
import math
import re

method_labels = {
    "sequence_matcher": "SequenceMatcher",
    "word_tfidf_cosine": "Word TF-IDF Cosine",
    "char_tfidf_cosine": "Character TF-IDF Cosine",
}

FLOAT_TOLERANCE = 1e-12

exact_results = [
    result
    for result in benchmark_results
    if result["label"] == "EXACT"
]

controlled_results = [
    result
    for result in benchmark_results
    if result["label"] == "CONTROLLED_NEAR"
]

independent_results = [
    result
    for result in benchmark_results
    if result["label"] == "RANDOM_INDEPENDENT"
]

pair_lookup = {
    (
        pair["label"],
        pair["left_id"],
        pair["right_id"],
    ): pair
    for pair in benchmark_pairs
}

def default_word_tokens(text):
    return re.findall(
        r"(?u)\b\w\w+\b",
        text,
    )

print("=== EXACT-PAIR METHOD INTEGRITY ===")

method_failures = {}

for method, label in method_labels.items():
    failures = [
        result
        for result in exact_results
        if not math.isclose(
            result[method],
            1.0,
            rel_tol=FLOAT_TOLERANCE,
            abs_tol=FLOAT_TOLERANCE,
        )
    ]

    method_failures[method] = failures

    print(f"\n{label}:")
    print(" Exact pairs:", len(exact_results))
    print(" Exact score ~= 1:", len(exact_results) - len(failures))
    print(" Exact integrity failures:", len(failures))

    for failure in failures:
        pair = pair_lookup[
            (
                failure["label"],
                failure["left_id"],
                failure["right_id"],
            )
        ]

        text = pair["left_text"]
        tokens = default_word_tokens(text)

        print("\n Failure:")
        print("  left_id:", failure["left_id"])
        print("  right_id:", failure["right_id"])
        print("  score:", failure[method])
        print("  characters:", len(text))
        print("  default word tokens:", len(tokens))
        print("  tokens:", tokens[:20])
        print("  text repr:", repr(text[:300]))

print("\n=== FLOATING-POINT RANGE VALIDATION ===")

strict_range_overshoots = []
true_range_failures = []

for result in benchmark_results:
    for method in method_labels:
        score = result[method]

        if not (0.0 <= score <= 1.0):
            strict_range_overshoots.append(
                {
                    "label": result["label"],
                    "left_id": result["left_id"],
                    "right_id": result["right_id"],
                    "method": method,
                    "score": score,
                }
            )

        if not (
            -FLOAT_TOLERANCE
            <= score
            <= 1.0 + FLOAT_TOLERANCE
        ):
            true_range_failures.append(
                {
                    "label": result["label"],
                    "left_id": result["left_id"],
                    "right_id": result["right_id"],
                    "method": method,
                    "score": score,
                }
            )

print(
    "Strict [0, 1] floating-point overshoots:",
    len(strict_range_overshoots),
)

if strict_range_overshoots:
    overshoot_values = [
        item["score"] - 1.0
        for item in strict_range_overshoots
        if item["score"] > 1.0
    ]

    undershoot_values = [
        abs(item["score"])
        for item in strict_range_overshoots
        if item["score"] < 0.0
    ]

    print(
        " Maximum positive overshoot:",
        max(overshoot_values) if overshoot_values else 0.0,
    )
    print(
        " Maximum negative undershoot:",
        max(undershoot_values) if undershoot_values else 0.0,
    )

print(
    "Scores outside tolerance-aware similarity range:",
    len(true_range_failures),
)

for failure in true_range_failures[:20]:
    print(failure)

print("\n=== CONTROLLED-NEAR VS RANDOM-INDEPENDENT SEPARATION ===")

separation_summary = {}

for method, label in method_labels.items():
    controlled_scores = [
        result[method]
        for result in controlled_results
    ]

    random_scores = [
        result[method]
        for result in independent_results
    ]

    controlled_min = min(controlled_scores)
    controlled_median = float(np.median(controlled_scores))
    random_max = max(random_scores)
    random_median = float(np.median(random_scores))

    observed_margin = controlled_min - random_max

    separation_summary[method] = {
        "controlled_min": controlled_min,
        "controlled_median": controlled_median,
        "random_max": random_max,
        "random_median": random_median,
        "observed_margin": observed_margin,
    }

    print(f"\n{label}:")
    print(
        " Controlled-near min:",
        round(controlled_min, 6),
    )
    print(
        " Controlled-near median:",
        round(controlled_median, 6),
    )
    print(
        " Random-independent max:",
        round(random_max, 6),
    )
    print(
        " Random-independent median:",
        round(random_median, 6),
    )
    print(
        " Observed benchmark margin:",
        round(observed_margin, 6),
    )

print("\n=== METHOD STABILITY SUMMARY ===")

for method, label in method_labels.items():
    exact_failure_count = len(method_failures[method])
    margin = separation_summary[method]["observed_margin"]

    print(
        f"{label}:",
        "| exact failures =", exact_failure_count,
        "| controlled/random observed margin =",
        round(margin, 6),
    )

sequence_exact_ok = (
    len(method_failures["sequence_matcher"]) == 0
)

char_exact_ok = (
    len(method_failures["char_tfidf_cosine"]) == 0
)

word_failures = method_failures["word_tfidf_cosine"]

word_failure_explained = (
    len(word_failures) > 0
    and all(
        len(
            default_word_tokens(
                pair_lookup[
                    (
                        failure["label"],
                        failure["left_id"],
                        failure["right_id"],
                    )
                ]["left_text"]
            )
        ) == 0
        for failure in word_failures
    )
)

score_range_ok = (
    len(true_range_failures) == 0
)

print("\nValidation:")
print(
    "SequenceMatcher exact integrity:",
    "PASS" if sequence_exact_ok else "FAIL",
)
print(
    "Character TF-IDF exact integrity:",
    "PASS" if char_exact_ok else "FAIL",
)
print(
    "Word TF-IDF failures explained by empty word tokenization:",
    "PASS" if word_failure_explained else "FAIL",
)
print(
    "Floating-point overshoots documented:",
    "PASS",
)
print(
    "Tolerance-aware similarity range:",
    "PASS" if score_range_ok else "FAIL",
)

cell_gate = (
    sequence_exact_ok
    and char_exact_ok
    and word_failure_explained
    and score_range_ok
)

print(
    "CODE CELL 12 GATE:",
    "PASS" if cell_gate else "FAIL",
)
```

<div style="direction: rtl; white-space: normal; line-height: 1.7;">

<h3>Benchmark بازیابی Candidate برای Near-duplicate درون هر Source</h3>

<p>
Benchmark قبلی Similarity Score را روی Pairهای از پیش تعیین‌شده بررسی کرد.
اما Near-duplicate detection در Dataset واقعی یک مسئله دوم نیز دارد:
ابتدا باید از میان هزاران Document، Candidateهای محتمل پیدا شوند و سپس آن Candidateها دقیق‌تر بررسی شوند.
</p>

<p>
به همین دلیل در این مرحله قابلیت
<code>Candidate Retrieval</code>
دو روش
<code>Word TF-IDF</code>
و
<code>Character TF-IDF</code>
آزمایش می‌شود.
<code>SequenceMatcher</code>
در این مرحله Candidate Generator نیست، زیرا مقایسه مستقیم همه Documentها با یکدیگر برای Corpus بزرگ روش مناسبی برای Candidate Generation اولیه محسوب نمی‌شود.
</p>

<p>
آزمایش به‌صورت Source-aware انجام می‌شود؛ یعنی Website فقط در برابر Website،
Book فقط در برابر Book
و Transcript فقط در برابر Transcript جست‌وجو می‌شود.
این تصمیم بر اساس Profiling واقعی مرحله قبل است که تفاوت شدید Granularity سه Source را اثبات کرد.
</p>

<h4>روش Benchmark</h4>

<ul>
<li>از Documentهای واقعی با متن حداقل <code>300</code> کاراکتر که در Benchmark قبلی انتخاب شده‌اند استفاده می‌شود.</li>
<li>نسخه Controlled Near همان Document به‌عنوان Query استفاده می‌شود.</li>
<li>تمام Documentهای واقعی همان Source به‌عنوان Search Corpus در نظر گرفته می‌شوند.</li>
<li>بررسی می‌شود Document اصلی در رتبه‌های <code>Top-1</code>، <code>Top-3</code>، <code>Top-5</code> و <code>Top-10</code> قرار می‌گیرد یا خیر.</li>
<li>این مقادیر فقط برای ارزیابی Retrieval هستند و هیچ‌کدام Threshold تولید محسوب نمی‌شوند.</li>
</ul>

<h4>چرا این آزمایش لازم است؟</h4>

<p>
ممکن است یک Similarity Metric برای یک Pair Score خوبی تولید کند، اما در Corpus واقعی Candidate درست را در میان تعداد زیادی Document مشابه پیدا نکند.
بنابراین پیش از اجرای Near-duplicate detection روی کل Dataset باید Retrieval Recall واقعی آن بررسی شود.
</p>

<h4>معیار PASS</h4>

<p>
برای تمام Queryهای Controlled Near باید Ranking بدون خطا تولید شود،
Document اصلی هر Query باید در Corpus همان Source وجود داشته باشد
و نتایج هر دو Candidate Generator به‌صورت کامل گزارش شوند.
بالا یا پایین بودن Recall خاصی در این مرحله شرط PASS نیست؛ نتیجه برای انتخاب روش مرحله بعد استفاده خواهد شد.
</p>

</div>

```python
# Code Cell 13

from collections import Counter, defaultdict
import json

import numpy as np
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# -------------------------------------------------------------------
# Rebuild the controlled-query set from Code Cell 11
# -------------------------------------------------------------------

controlled_queries = []

for pair in benchmark_pairs:
    if pair["label"] != "CONTROLLED_NEAR":
        continue

    controlled_queries.append(
        {
            "target_document_id": pair["left_id"],
            "query_id": pair["right_id"],
            "query_text": pair["right_text"],
        }
    )

# -------------------------------------------------------------------
# Build independent corpus records from disk
# -------------------------------------------------------------------

retrieval_corpus = []

with input_path.open("r", encoding="utf-8") as handle:
    for line in handle:
        record = json.loads(line)

        retrieval_corpus.append(
            {
                "document_id": record["document_id"],
                "source_type": record["source_type"],
                "source_id": record["source_id"],
                "text": record["text"].strip(),
            }
        )

record_by_id = {
    record["document_id"]: record
    for record in retrieval_corpus
}

# Attach source type to each controlled query.
for query in controlled_queries:
    target_record = record_by_id.get(
        query["target_document_id"]
    )

    query["source_type"] = (
        target_record["source_type"]
        if target_record is not None
        else None
    )

missing_targets = [
    query["target_document_id"]
    for query in controlled_queries
    if query["target_document_id"] not in record_by_id
]

print("=== SOURCE-AWARE CANDIDATE RETRIEVAL BENCHMARK ===")
print("Controlled queries:", len(controlled_queries))
print("Missing target documents:", len(missing_targets))

query_distribution = Counter(
    query["source_type"]
    for query in controlled_queries
)

print("\nQuery distribution:")
for source_type in ["website", "book", "transcript"]:
    print(
        f"- {source_type}:",
        query_distribution.get(source_type, 0),
    )

# -------------------------------------------------------------------
# Candidate-generator configurations
# -------------------------------------------------------------------

generator_configs = {
    "WORD_TFIDF": {
        "analyzer": "word",
        "ngram_range": (1, 2),
    },
    "CHAR_TFIDF": {
        "analyzer": "char_wb",
        "ngram_range": (3, 5),
    },
}

retrieval_results = []

# -------------------------------------------------------------------
# Source-aware retrieval
# -------------------------------------------------------------------

for source_type in ["website", "book", "transcript"]:
    source_records = [
        record
        for record in retrieval_corpus
        if record["source_type"] == source_type
    ]

    source_queries = [
        query
        for query in controlled_queries
        if query["source_type"] == source_type
    ]

    if not source_queries:
        continue

    corpus_texts = [
        record["text"]
        for record in source_records
    ]

    corpus_ids = [
        record["document_id"]
        for record in source_records
    ]

    id_to_index = {
        document_id: index
        for index, document_id in enumerate(corpus_ids)
    }

    print("\n" + "=" * 80)
    print("SOURCE:", source_type)
    print("Corpus documents:", len(source_records))
    print("Benchmark queries:", len(source_queries))

    for generator_name, config in generator_configs.items():
        vectorizer = TfidfVectorizer(
            analyzer=config["analyzer"],
            ngram_range=config["ngram_range"],
            lowercase=False,
            min_df=1,
        )

        corpus_matrix = vectorizer.fit_transform(
            corpus_texts
        )

        query_matrix = vectorizer.transform(
            [
                query["query_text"]
                for query in source_queries
            ]
        )

        similarity_matrix = cosine_similarity(
            query_matrix,
            corpus_matrix,
        )

        print(f"\nGenerator: {generator_name}")
        print(
            "Vocabulary size:",
            len(vectorizer.vocabulary_),
        )

        for query_index, query in enumerate(source_queries):
            target_id = query["target_document_id"]
            target_index = id_to_index[target_id]

            scores = similarity_matrix[query_index]

            ranking = np.argsort(
                -scores,
                kind="stable",
            )

            target_rank = int(
                np.where(ranking == target_index)[0][0]
            ) + 1

            target_score = float(
                scores[target_index]
            )

            top_candidate_index = int(
                ranking[0]
            )

            top_candidate_id = corpus_ids[
                top_candidate_index
            ]

            top_candidate_score = float(
                scores[top_candidate_index]
            )

            retrieval_results.append(
                {
                    "generator": generator_name,
                    "source_type": source_type,
                    "query_id": query["query_id"],
                    "target_document_id": target_id,
                    "target_rank": target_rank,
                    "target_score": target_score,
                    "top_candidate_id": top_candidate_id,
                    "top_candidate_score": top_candidate_score,
                }
            )

# -------------------------------------------------------------------
# Retrieval metrics
# -------------------------------------------------------------------

def recall_at_k(results, k):
    if not results:
        return None

    hits = sum(
        result["target_rank"] <= k
        for result in results
    )

    return hits / len(results)

print("\n=== RETRIEVAL RECALL SUMMARY ===")

for generator_name in generator_configs:
    generator_results = [
        result
        for result in retrieval_results
        if result["generator"] == generator_name
    ]

    print("\nGENERATOR:", generator_name)
    print("Queries:", len(generator_results))

    for k in [1, 3, 5, 10]:
        value = recall_at_k(
            generator_results,
            k,
        )

        print(
            f" Recall@{k}:",
            round(value, 6)
            if value is not None
            else None,
        )

    ranks = [
        result["target_rank"]
        for result in generator_results
    ]

    print(
        " Median target rank:",
        float(np.median(ranks))
        if ranks
        else None,
    )
    print(
        " Worst target rank:",
        max(ranks)
        if ranks
        else None,
    )

    print("\nBy source:")

    for source_type in [
        "website",
        "book",
        "transcript",
    ]:
        source_results = [
            result
            for result in generator_results
            if result["source_type"] == source_type
        ]

        if not source_results:
            continue

        print(f" {source_type}:")

        for k in [1, 3, 5, 10]:
            value = recall_at_k(
                source_results,
                k,
            )

            print(
                f"  Recall@{k}:",
                round(value, 6),
            )

        source_ranks = [
            result["target_rank"]
            for result in source_results
        ]

        print(
            "  Median rank:",
            float(np.median(source_ranks)),
        )
        print(
            "  Worst rank:",
            max(source_ranks),
        )

# -------------------------------------------------------------------
# Inspect retrieval misses
# -------------------------------------------------------------------

print("\n=== TOP-1 MISSES ===")

for generator_name in generator_configs:
    misses = [
        result
        for result in retrieval_results
        if result["generator"] == generator_name
        and result["target_rank"] > 1
    ]

    print(
        f"\n{generator_name} Top-1 misses:",
        len(misses),
    )

    misses.sort(
        key=lambda item: (
            -item["target_rank"],
            item["target_document_id"],
        )
    )

    for miss in misses[:20]:
        print(
            "- source:",
            miss["source_type"],
        )
        print(
            "  target:",
            miss["target_document_id"],
        )
        print(
            "  target rank:",
            miss["target_rank"],
        )
        print(
            "  target score:",
            round(
                miss["target_score"],
                6,
            ),
        )
        print(
            "  top candidate:",
            miss["top_candidate_id"],
        )
        print(
            "  top candidate score:",
            round(
                miss["top_candidate_score"],
                6,
            ),
        )

# -------------------------------------------------------------------
# Validation
# -------------------------------------------------------------------

expected_result_count = (
    len(controlled_queries)
    * len(generator_configs)
)

result_count_ok = (
    len(retrieval_results)
    == expected_result_count
)

all_targets_ranked = all(
    isinstance(result["target_rank"], int)
    and result["target_rank"] >= 1
    for result in retrieval_results
)

all_scores_finite = all(
    np.isfinite(
        result["target_score"]
    )
    and np.isfinite(
        result["top_candidate_score"]
    )
    for result in retrieval_results
)

print("\nValidation:")
print(
    "All controlled targets found in corpus:",
    "PASS" if not missing_targets else "FAIL",
)
print(
    "Expected retrieval result count:",
    "PASS" if result_count_ok else "FAIL",
)
print(
    "All targets ranked:",
    "PASS" if all_targets_ranked else "FAIL",
)
print(
    "All retrieval scores finite:",
    "PASS" if all_scores_finite else "FAIL",
)

cell_gate = (
    not missing_targets
    and result_count_ok
    and all_targets_ranked
    and all_scores_finite
)

print(
    "CODE CELL 13 GATE:",
    "PASS" if cell_gate else "FAIL",
)
```

<div style="direction: rtl; white-space: normal; line-height: 1.7;">

<h3>Discovery نزدیک‌ترین همسایه‌های غیرExact در Corpus واقعی</h3>

<p>
Benchmarkهای کنترل‌شده نشان دادند که Character TF-IDF از نظر Exact integrity پایدار است و در Candidate Retrieval نیز Document اصلی را برای تمام Queryهای آزمایشی در رتبه اول بازیابی کرده است.
</p>

<p>
اکنون باید رفتار همین Candidate Generator روی Dataset واقعی بررسی شود.
در این مرحله برای هر Document، نزدیک‌ترین Documentهای همان Source جست‌وجو می‌شوند و تمام اعضایی که
<code>text_sha256</code>
یکسان دارند از تحلیل Near-duplicate کنار گذاشته می‌شوند.
</p>

<p>
حذف Exact match از این تحلیل ضروری است، زیرا Exact Duplicate قبلاً جداگانه اندازه‌گیری شده و هدف این مرحله پیدا کردن شباهت‌های
<strong>غیرExact</strong>
است.
</p>

<p>
این مرحله هنوز Threshold تولید تعریف نمی‌کند.
Similarity Scoreها فقط اندازه‌گیری و توزیع آن‌ها بررسی می‌شود تا مشخص شود Corpus واقعی چه ساختاری دارد و آیا خوشه‌ای طبیعی از Candidateهای بسیار مشابه وجود دارد یا خیر.
</p>

<h4>روش Candidate Generation</h4>

<ul>
<li>جست‌وجو برای Website، Book و Transcript جداگانه انجام می‌شود.</li>
<li>از Character TF-IDF با همان تنظیمات Benchmark قبلی استفاده می‌شود.</li>
<li>برای هر Document چند همسایه نزدیک بازیابی می‌شود.</li>
<li>خود Document و تمام Exact Duplicateهای دارای <code>text_sha256</code> یکسان کنار گذاشته می‌شوند.</li>
<li>نزدیک‌ترین Candidate غیرExact برای هر Document ثبت می‌شود.</li>
</ul>

<h4>خروجی این مرحله</h4>

<ul>
<li>تعداد Documentهایی که Candidate غیرExact معتبر دارند.</li>
<li>توزیع Score نزدیک‌ترین Candidate به تفکیک Source.</li>
<li>Quantileهای Similarity در Corpus واقعی.</li>
<li>بالاترین Candidate pairهای غیرExact برای Inspection انسانی.</li>
<li>Text preview، Document ID، Source ID و Score هر Pair.</li>
</ul>

<h4>اصل مهم</h4>

<p>
Score بالا در این مرحله به‌تنهایی به معنی Duplicate نیست.
ممکن است دو متن نسخه‌های نزدیک،
Boilerplateهای کمی متفاوت،
Headingهای مشابه،
جملات قالبی،
یا محتوای واقعاً تکرارشده باشند.
Classification نهایی فقط بعد از Inspection شواهد انجام می‌شود.
</p>

<h4>معیار PASS</h4>

<p>
برای هر سه Source باید Candidate Generation بدون خطا اجرا شود،
Exact matchها نباید به‌عنوان Near-duplicate وارد خروجی شوند،
تمام Scoreها باید finite باشند
و Pairهای گزارش‌شده باید واقعاً دارای <code>text_sha256</code> متفاوت باشند.
</p>

</div>

```python
# Code Cell 14

from collections import defaultdict
import json
import math

import numpy as np
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.neighbors import NearestNeighbors

NEIGHBORS_TO_REQUEST = 12

# -------------------------------------------------------------------
# Reload normalized corpus independently
# -------------------------------------------------------------------

real_corpus_records = []

with input_path.open("r", encoding="utf-8") as handle:
    for line in handle:
        record = json.loads(line)

        real_corpus_records.append(
            {
                "document_id": record["document_id"],
                "source_type": record["source_type"],
                "source_id": record["source_id"],
                "text_sha256": record["text_sha256"],
                "text": record["text"].strip(),
                "title": record.get("title"),
            }
        )

nearest_nonexact_results = []

# -------------------------------------------------------------------
# Source-aware Character TF-IDF nearest-neighbor discovery
# -------------------------------------------------------------------

for source_type in ["website", "book", "transcript"]:
    source_records = [
        record
        for record in real_corpus_records
        if record["source_type"] == source_type
    ]

    texts = [
        record["text"]
        for record in source_records
    ]

    print("\n" + "=" * 80)
    print("SOURCE:", source_type)
    print("Documents:", len(source_records))

    vectorizer = TfidfVectorizer(
        analyzer="char_wb",
        ngram_range=(3, 5),
        lowercase=False,
        min_df=1,
    )

    matrix = vectorizer.fit_transform(texts)

    print(
        "TF-IDF matrix shape:",
        matrix.shape,
    )

    neighbor_count = min(
        NEIGHBORS_TO_REQUEST,
        len(source_records),
    )

    nn_model = NearestNeighbors(
        n_neighbors=neighbor_count,
        metric="cosine",
        algorithm="brute",
        n_jobs=-1,
    )

    nn_model.fit(matrix)

    distances, indices = nn_model.kneighbors(
        matrix,
        return_distance=True,
    )

    found_count = 0

    for row_index, record in enumerate(source_records):
        selected = None

        for distance, neighbor_index in zip(
            distances[row_index],
            indices[row_index],
        ):
            neighbor = source_records[
                int(neighbor_index)
            ]

            # Exclude self.
            if (
                neighbor["document_id"]
                == record["document_id"]
            ):
                continue

            # Exclude all exact-text duplicates.
            if (
                neighbor["text_sha256"]
                == record["text_sha256"]
            ):
                continue

            similarity = 1.0 - float(distance)

            selected = {
                "source_type": source_type,
                "document_id": record["document_id"],
                "source_id": record["source_id"],
                "title": record["title"],
                "text_sha256": record["text_sha256"],
                "text": record["text"],
                "candidate_document_id": neighbor["document_id"],
                "candidate_source_id": neighbor["source_id"],
                "candidate_title": neighbor["title"],
                "candidate_text_sha256": neighbor["text_sha256"],
                "candidate_text": neighbor["text"],
                "similarity": similarity,
            }

            break

        if selected is not None:
            nearest_nonexact_results.append(
                selected
            )
            found_count += 1

    print(
        "Nearest non-exact candidates found:",
        found_count,
    )

# -------------------------------------------------------------------
# Distribution summary
# -------------------------------------------------------------------

print("\n=== REAL-CORPUS NEAREST NON-EXACT SIMILARITY ===")

source_similarity_values = {}

for source_type in [
    "website",
    "book",
    "transcript",
]:
    values = [
        result["similarity"]
        for result in nearest_nonexact_results
        if result["source_type"] == source_type
    ]

    source_similarity_values[source_type] = values

    print("\nSOURCE:", source_type)
    print("Documents with candidate:", len(values))

    if not values:
        continue

    array = np.asarray(
        values,
        dtype=float,
    )

    print(
        " Min:",
        round(float(np.min(array)), 6),
    )
    print(
        " P25:",
        round(float(np.percentile(array, 25)), 6),
    )
    print(
        " Median:",
        round(float(np.median(array)), 6),
    )
    print(
        " P75:",
        round(float(np.percentile(array, 75)), 6),
    )
    print(
        " P90:",
        round(float(np.percentile(array, 90)), 6),
    )
    print(
        " P95:",
        round(float(np.percentile(array, 95)), 6),
    )
    print(
        " P99:",
        round(float(np.percentile(array, 99)), 6),
    )
    print(
        " Max:",
        round(float(np.max(array)), 6),
    )

# -------------------------------------------------------------------
# Convert directed nearest-neighbor results into unique unordered pairs
# -------------------------------------------------------------------

unique_pairs = {}

for result in nearest_nonexact_results:
    pair_key = tuple(
        sorted(
            [
                result["document_id"],
                result["candidate_document_id"],
            ]
        )
    )

    existing = unique_pairs.get(pair_key)

    if (
        existing is None
        or result["similarity"]
        > existing["similarity"]
    ):
        unique_pairs[pair_key] = result

unique_pair_results = list(
    unique_pairs.values()
)

unique_pair_results.sort(
    key=lambda item: (
        -item["similarity"],
        item["source_type"],
        item["document_id"],
        item["candidate_document_id"],
    )
)

print("\n=== HIGHEST REAL NON-EXACT CANDIDATE PAIRS ===")
print(
    "Unique nearest-neighbor pairs:",
    len(unique_pair_results),
)

for index, result in enumerate(
    unique_pair_results[:60],
    start=1,
):
    print("\n" + "-" * 100)
    print("PAIR", index)
    print(
        "Source:",
        result["source_type"],
    )
    print(
        "Similarity:",
        round(
            result["similarity"],
            6,
        ),
    )

    print(
        "A:",
        result["document_id"],
    )
    print(
        "A source_id:",
        result["source_id"],
    )
    print(
        "A title:",
        result["title"],
    )

    print(
        "B:",
        result["candidate_document_id"],
    )
    print(
        "B source_id:",
        result["candidate_source_id"],
    )
    print(
        "B title:",
        result["candidate_title"],
    )

    print(
        "Same source_id:",
        result["source_id"]
        == result["candidate_source_id"],
    )

    print(
        "Same text_sha256:",
        result["text_sha256"]
        == result["candidate_text_sha256"],
    )

    preview_a = (
        result["text"]
        .replace("\n", " ")
        .strip()
    )

    preview_b = (
        result["candidate_text"]
        .replace("\n", " ")
        .strip()
    )

    print(
        "A text:",
        preview_a[:500],
    )
    print(
        "B text:",
        preview_b[:500],
    )

# -------------------------------------------------------------------
# Validation
# -------------------------------------------------------------------

expected_source_counts = {
    "website": 529,
    "book": 2701,
    "transcript": 16226,
}

candidate_counts = {
    source_type: sum(
        result["source_type"] == source_type
        for result in nearest_nonexact_results
    )
    for source_type in expected_source_counts
}

all_scores_finite = all(
    math.isfinite(
        result["similarity"]
    )
    for result in nearest_nonexact_results
)

no_exact_pairs = all(
    result["text_sha256"]
    != result["candidate_text_sha256"]
    for result in nearest_nonexact_results
)

no_self_pairs = all(
    result["document_id"]
    != result["candidate_document_id"]
    for result in nearest_nonexact_results
)

all_sources_have_candidates = all(
    candidate_counts[source_type] > 0
    for source_type in expected_source_counts
)

print("\nValidation:")

for source_type in [
    "website",
    "book",
    "transcript",
]:
    print(
        f"{source_type} candidate records:",
        candidate_counts[source_type],
    )

print(
    "All sources produced candidates:",
    "PASS" if all_sources_have_candidates else "FAIL",
)
print(
    "All scores finite:",
    "PASS" if all_scores_finite else "FAIL",
)
print(
    "Self-pairs excluded:",
    "PASS" if no_self_pairs else "FAIL",
)
print(
    "Exact-text pairs excluded:",
    "PASS" if no_exact_pairs else "FAIL",
)

cell_gate = (
    all_sources_have_candidates
    and all_scores_finite
    and no_self_pairs
    and no_exact_pairs
)

print(
    "CODE CELL 14 GATE:",
    "PASS" if cell_gate else "FAIL",
)
```

<div style="direction: rtl; white-space: normal; line-height: 1.7;">

<h3>تحلیل چندمعیاره Candidateهای Near-duplicate</h3>

<p>
در مرحله قبل Character TF-IDF برای هر Document نزدیک‌ترین Candidate غیرExact را پیدا کرد.
خروجی نشان داد که Similarity بالا به‌تنهایی برای تشخیص Near-duplicate کافی نیست.
</p>

<p>
برای مثال چند نوع متفاوت از شباهت در Candidateهای با Score بالا مشاهده شد:
</p>

<ul>
<li>Boilerplateهای تقریباً یکسان برنامه‌های رادیویی.</li>
<li>عبارت‌های کوتاه با یک یا دو واژه اضافه یا حذف‌شده.</li>
<li>متن‌هایی با همان کلمات ولی ترتیب متفاوت.</li>
<li>رشته‌های کم‌اطلاعات مانند نقطه‌چین‌های کاربرگ.</li>
<li>پاراگراف‌هایی که احتمالاً نسخه بسیار نزدیک یکدیگر هستند.</li>
</ul>

<p>
بنابراین در این مرحله Candidateها فقط با Character TF-IDF قضاوت نمی‌شوند.
برای هر Pair چند ویژگی مستقل دیگر محاسبه می‌شود تا ماهیت شباهت بهتر شناخته شود.
</p>

<h4>ویژگی‌های تکمیلی</h4>

<ul>
<li><code>Character TF-IDF similarity</code>: Score مرحله Candidate Generation.</li>
<li><code>SequenceMatcher</code>: میزان شباهت ترتیبی دو رشته.</li>
<li><code>Token Jaccard</code>: میزان اشتراک مجموعه واژه‌های دو متن.</li>
<li><code>Length Ratio</code>: میزان نزدیکی طول دو متن.</li>
<li>تعداد Character و Word هر طرف.</li>
<li>تعداد Characterهای حرفی/عددی برای شناسایی متن‌های کم‌اطلاعات.</li>
<li>نسبت Characterهای حرفی/عددی به کل متن.</li>
<li>یکسان یا متفاوت بودن <code>source_id</code>.</li>
</ul>

<p>
این ویژگی‌ها هنوز Rule حذف یا Quality Policy نیستند.
هدف فقط ایجاد Evidence چندبعدی برای تفکیک انواع Candidate است.
</p>

<h4>Similarity Bandها</h4>

<p>
برای مشاهده توزیع Candidateها چند بازه توصیفی ساخته می‌شود.
این بازه‌ها صرفاً ابزار Profiling هستند و Threshold نهایی Near-duplicate محسوب نمی‌شوند.
</p>

<ul>
<li><code>0.98_PLUS</code></li>
<li><code>0.95_TO_0.98</code></li>
<li><code>0.90_TO_0.95</code></li>
<li><code>0.80_TO_0.90</code></li>
<li><code>BELOW_0.80</code></li>
</ul>

<h4>هدف مرحله</h4>

<p>
باید مشخص شود Candidateهای دارای Character Similarity بسیار بالا از نظر Sequence، Token overlap، طول و مقدار محتوای واقعی چه تفاوتی با یکدیگر دارند.
این Evidence در مرحله بعد برای طراحی Classification و Near-duplicate Policy استفاده خواهد شد.
</p>

<h4>معیار PASS</h4>

<p>
تمام Pairهای یکتای مرحله قبل باید Featureهای تکمیلی دریافت کنند،
هیچ Score غیرFinite نباید وجود داشته باشد،
هیچ Exact Duplicate نباید وارد این مجموعه شود
و مجموع Pairهای تمام Similarity Bandها باید با کل Candidateهای یکتا برابر باشد.
</p>

</div>

```python
# Code Cell 15

from collections import Counter, defaultdict
from difflib import SequenceMatcher
import math
import re

def token_set(text):
    return set(
        re.findall(
            r"[\w\u0600-\u06FF\u200c]+",
            text,
            flags=re.UNICODE,
        )
    )

def token_jaccard(text_a, text_b):
    tokens_a = token_set(text_a)
    tokens_b = token_set(text_b)

    union = tokens_a | tokens_b

    if not union:
        return 1.0 if text_a == text_b else 0.0

    return len(tokens_a & tokens_b) / len(union)

def length_ratio(text_a, text_b):
    len_a = len(text_a)
    len_b = len(text_b)

    if len_a == 0 and len_b == 0:
        return 1.0

    if max(len_a, len_b) == 0:
        return 0.0

    return min(len_a, len_b) / max(len_a, len_b)

def informative_char_count(text):
    return sum(
        char.isalnum()
        for char in text
    )

def informative_char_ratio(text):
    if not text:
        return 0.0

    return informative_char_count(text) / len(text)

def similarity_band(score):
    if score >= 0.98:
        return "0.98_PLUS"
    if score >= 0.95:
        return "0.95_TO_0.98"
    if score >= 0.90:
        return "0.90_TO_0.95"
    if score >= 0.80:
        return "0.80_TO_0.90"
    return "BELOW_0.80"

pair_feature_records = []

for result in unique_pair_results:
    text_a = result["text"]
    text_b = result["candidate_text"]

    sequence_score = SequenceMatcher(
        None,
        text_a,
        text_b,
        autojunk=False,
    ).ratio()

    jaccard_score = token_jaccard(
        text_a,
        text_b,
    )

    len_ratio = length_ratio(
        text_a,
        text_b,
    )

    chars_a = len(text_a)
    chars_b = len(text_b)

    words_a = len(text_a.split())
    words_b = len(text_b.split())

    informative_a = informative_char_count(
        text_a
    )

    informative_b = informative_char_count(
        text_b
    )

    informative_ratio_a = informative_char_ratio(
        text_a
    )

    informative_ratio_b = informative_char_ratio(
        text_b
    )

    pair_feature_records.append(
        {
            **result,
            "similarity_band": similarity_band(
                result["similarity"]
            ),
            "sequence_matcher": sequence_score,
            "token_jaccard": jaccard_score,
            "length_ratio": len_ratio,
            "chars_a": chars_a,
            "chars_b": chars_b,
            "words_a": words_a,
            "words_b": words_b,
            "informative_chars_a": informative_a,
            "informative_chars_b": informative_b,
            "informative_ratio_a": informative_ratio_a,
            "informative_ratio_b": informative_ratio_b,
        }
    )

print("=== MULTI-METRIC NEAR-DUPLICATE CANDIDATE PROFILE ===")
print(
    "Unique candidate pairs:",
    len(pair_feature_records),
)

# -------------------------------------------------------------------
# Band distribution
# -------------------------------------------------------------------

band_order = [
    "0.98_PLUS",
    "0.95_TO_0.98",
    "0.90_TO_0.95",
    "0.80_TO_0.90",
    "BELOW_0.80",
]

band_counts = Counter(
    row["similarity_band"]
    for row in pair_feature_records
)

print("\n=== SIMILARITY BAND DISTRIBUTION ===")

for band in band_order:
    print(
        f"{band}:",
        band_counts.get(band, 0),
    )

print("\n=== BAND DISTRIBUTION BY SOURCE ===")

for source_type in [
    "website",
    "book",
    "transcript",
]:
    print("\nSOURCE:", source_type)

    source_rows = [
        row
        for row in pair_feature_records
        if row["source_type"] == source_type
    ]

    for band in band_order:
        count = sum(
            row["similarity_band"] == band
            for row in source_rows
        )

        print(
            f"- {band}:",
            count,
        )

# -------------------------------------------------------------------
# Descriptive metric summary by band
# -------------------------------------------------------------------

print("\n=== MULTI-METRIC SUMMARY BY BAND ===")

for band in band_order:
    rows = [
        row
        for row in pair_feature_records
        if row["similarity_band"] == band
    ]

    if not rows:
        continue

    print("\nBAND:", band)
    print("Pairs:", len(rows))

    for metric in [
        "similarity",
        "sequence_matcher",
        "token_jaccard",
        "length_ratio",
    ]:
        values = np.asarray(
            [
                row[metric]
                for row in rows
            ],
            dtype=float,
        )

        print(f"\n{metric}:")
        print(
            " Min:",
            round(
                float(np.min(values)),
                6,
            ),
        )
        print(
            " Median:",
            round(
                float(np.median(values)),
                6,
            ),
        )
        print(
            " P90:",
            round(
                float(
                    np.percentile(
                        values,
                        90,
                    )
                ),
                6,
            ),
        )
        print(
            " Max:",
            round(
                float(np.max(values)),
                6,
            ),
        )

# -------------------------------------------------------------------
# Inspect top high-similarity pairs with all features
# -------------------------------------------------------------------

high_similarity_rows = [
    row
    for row in pair_feature_records
    if row["similarity"] >= 0.95
]

high_similarity_rows.sort(
    key=lambda row: (
        -row["similarity"],
        -row["sequence_matcher"],
        row["source_type"],
        row["document_id"],
    )
)

print("\n=== HIGH-SIMILARITY MULTI-METRIC INSPECTION ===")
print(
    "Pairs with Character TF-IDF >= 0.95:",
    len(high_similarity_rows),
)

for index, row in enumerate(
    high_similarity_rows[:80],
    start=1,
):
    print("\n" + "-" * 100)
    print("PAIR", index)
    print(
        "Source:",
        row["source_type"],
    )
    print(
        "Band:",
        row["similarity_band"],
    )

    print(
        "Character TF-IDF:",
        round(
            row["similarity"],
            6,
        ),
    )
    print(
        "SequenceMatcher:",
        round(
            row["sequence_matcher"],
            6,
        ),
    )
    print(
        "Token Jaccard:",
        round(
            row["token_jaccard"],
            6,
        ),
    )
    print(
        "Length ratio:",
        round(
            row["length_ratio"],
            6,
        ),
    )

    print(
        "Same source_id:",
        row["source_id"]
        == row["candidate_source_id"],
    )

    print(
        "A chars/words/informative:",
        row["chars_a"],
        "/",
        row["words_a"],
        "/",
        row["informative_chars_a"],
    )

    print(
        "B chars/words/informative:",
        row["chars_b"],
        "/",
        row["words_b"],
        "/",
        row["informative_chars_b"],
    )

    print(
        "A informative ratio:",
        round(
            row["informative_ratio_a"],
            6,
        ),
    )

    print(
        "B informative ratio:",
        round(
            row["informative_ratio_b"],
            6,
        ),
    )

    print(
        "A:",
        row["document_id"],
    )
    print(
        "B:",
        row["candidate_document_id"],
    )

    print(
        "A text:",
        row["text"]
        .replace("\n", " ")[:500],
    )

    print(
        "B text:",
        row["candidate_text"]
        .replace("\n", " ")[:500],
    )

# -------------------------------------------------------------------
# Metric disagreement discovery
# -------------------------------------------------------------------

metric_disagreement_rows = sorted(
    pair_feature_records,
    key=lambda row: (
        -abs(
            row["similarity"]
            - row["sequence_matcher"]
        ),
        row["document_id"],
    ),
)

print("\n=== LARGEST CHARACTER/SEQUENCE DISAGREEMENTS ===")

for index, row in enumerate(
    metric_disagreement_rows[:40],
    start=1,
):
    difference = abs(
        row["similarity"]
        - row["sequence_matcher"]
    )

    print("\nPAIR", index)
    print(
        "Source:",
        row["source_type"],
    )
    print(
        "Character TF-IDF:",
        round(
            row["similarity"],
            6,
        ),
    )
    print(
        "SequenceMatcher:",
        round(
            row["sequence_matcher"],
            6,
        ),
    )
    print(
        "Absolute difference:",
        round(
            difference,
            6,
        ),
    )
    print(
        "Token Jaccard:",
        round(
            row["token_jaccard"],
            6,
        ),
    )
    print(
        "A:",
        row["text"]
        .replace("\n", " ")[:300],
    )
    print(
        "B:",
        row["candidate_text"]
        .replace("\n", " ")[:300],
    )

# -------------------------------------------------------------------
# Validation
# -------------------------------------------------------------------

all_features_finite = all(
    math.isfinite(
        row[metric]
    )
    for row in pair_feature_records
    for metric in [
        "similarity",
        "sequence_matcher",
        "token_jaccard",
        "length_ratio",
        "informative_ratio_a",
        "informative_ratio_b",
    ]
)

no_exact_pairs = all(
    row["text_sha256"]
    != row["candidate_text_sha256"]
    for row in pair_feature_records
)

band_accounting_ok = (
    sum(
        band_counts.values()
    )
    == len(pair_feature_records)
)

pair_count_preserved = (
    len(pair_feature_records)
    == len(unique_pair_results)
)

print("\nValidation:")
print(
    "Pair count preserved:",
    "PASS" if pair_count_preserved else "FAIL",
)
print(
    "All features finite:",
    "PASS" if all_features_finite else "FAIL",
)
print(
    "Exact-text pairs excluded:",
    "PASS" if no_exact_pairs else "FAIL",
)
print(
    "Similarity-band accounting:",
    "PASS" if band_accounting_ok else "FAIL",
)

cell_gate = (
    pair_count_preserved
    and all_features_finite
    and no_exact_pairs
    and band_accounting_ok
)

print(
    "CODE CELL 15 GATE:",
    "PASS" if cell_gate else "FAIL",
)
```

<div style="direction: rtl; white-space: normal; line-height: 1.7;">

<h3>تحلیل Context و رابطه ساختاری Candidateهای با شباهت بالا</h3>

<p>
تحلیل چندمعیاره نشان داد که Candidateهای با Similarity بالا از یک نوع واحد نیستند.
برای تصمیم Deduplication باید علاوه بر متن، جایگاه هر Document در Source نیز بررسی شود.
</p>

<p>
برای مثال دو Segment بسیار مشابه ممکن است:
در یک Episode و نزدیک به هم باشند،
در Episodeهای مختلف یک Program تکرار شده باشند،
در دو Book متفاوت قرار گرفته باشند،
یا فقط به دلیل ساختارهای کم‌اطلاعات مانند نقطه‌چین Similarity بالایی گرفته باشند.
</p>

<p>
در این مرحله هیچ Candidate حذف یا Merge نمی‌شود.
هدف فقط افزودن Context واقعی Source و ساختن Relationهای توصیفی است.
</p>

<h4>Contextهای مورد بررسی</h4>

<ul>
<li>یکسان یا متفاوت بودن <code>source_id</code>.</li>
<li>برای Book: یکسان بودن Work و Edition.</li>
<li>برای Transcript: یکسان بودن Program و Episode.</li>
<li>فاصله Utteranceها در یک Episode.</li>
<li>فاصله Paragraphها در یک Book edition.</li>
<li>وجود رابطه Containment؛ یعنی اینکه متن کوتاه‌تر به‌طور کامل داخل متن بلندتر باشد.</li>
<li>یکسان بودن مجموعه Tokenها با وجود تفاوت ترتیب.</li>
<li>وجود یا نبود Character حرفی/عددی واقعی.</li>
</ul>

<h4>Relationهای توصیفی</h4>

<p>
برای این مرحله چند Relation صرفاً تشخیصی ساخته می‌شود:
</p>

<ul>
<li><code>SAME_SOURCE_LOCAL</code>: دو رکورد در یک Source قرار دارند و فاصله ساختاری آن‌ها قابل محاسبه است.</li>
<li><code>SAME_SOURCE_NONLOCAL</code>: Source یکسان است اما فاصله ساختاری زیاد یا نامشخص است.</li>
<li><code>SAME_PROGRAM_DIFFERENT_EPISODE</code>: Transcriptهای یک Program در Episodeهای مختلف.</li>
<li><code>SAME_WORK_DIFFERENT_SOURCE</code>: واحدهای Book متعلق به یک Work اما Source ID متفاوت.</li>
<li><code>DIFFERENT_SOURCE_ID</code>: Source IDها متفاوت‌اند و Relation خاص‌تری اثبات نشده است.</li>
</ul>

<p>
این Relationها Policy نهایی نیستند و صرفاً Evidence برای مرحله تصمیم‌گیری هستند.
</p>

<h4>چرا این مرحله مهم است؟</h4>

<p>
اگر یک جمله در دو Utterance پشت سر هم تکرار شده باشد، احتمال تکرار گفتاری مطرح است.
اگر یک متن تقریباً ثابت در ده‌ها Episode یک Program دیده شود، احتمال Boilerplate مطرح می‌شود.
اگر دو Paragraph در دو کتاب متفاوت بسیار نزدیک باشند، مسئله Version یا Reuse محتوا مطرح می‌شود.
این حالات نباید با یک Rule واحد مدیریت شوند.
</p>

<h4>معیار PASS</h4>

<p>
تمام Candidateهای دارای Character TF-IDF حداقل <code>0.90</code> باید Context ساختاری دریافت کنند،
هیچ Document نباید در Lookup گم شود،
و Relationهای تشخیصی باید برای همه Pairها قابل محاسبه باشند.
</p>

</div>

```python
# Code Cell 16

from collections import Counter, defaultdict
import json
import re

# -------------------------------------------------------------------
# High-interest candidate population
# Uses an existing profiling band boundary, not a new production rule.
# -------------------------------------------------------------------

context_candidate_rows = [
    row
    for row in pair_feature_records
    if row["similarity"] >= 0.90
]

target_ids = {
    document_id
    for row in context_candidate_rows
    for document_id in [
        row["document_id"],
        row["candidate_document_id"],
    ]
}

# -------------------------------------------------------------------
# Reload full records independently from disk
# -------------------------------------------------------------------

context_record_lookup = {}

with input_path.open("r", encoding="utf-8") as handle:
    for line in handle:
        record = json.loads(line)

        if record["document_id"] in target_ids:
            context_record_lookup[
                record["document_id"]
            ] = record

missing_context_ids = sorted(
    target_ids
    - set(context_record_lookup)
)

# -------------------------------------------------------------------
# Structural helpers
# -------------------------------------------------------------------

def get_book_context(record):
    metadata = record.get("source_metadata")
    locator = record.get("locator")

    metadata = (
        metadata
        if isinstance(metadata, dict)
        else {}
    )

    locator = (
        locator
        if isinstance(locator, dict)
        else {}
    )

    return {
        "work_id": (
            metadata.get("work_id")
            or locator.get("work_id")
        ),
        "edition_id": (
            metadata.get("edition_id")
            or locator.get("edition_id")
        ),
        "paragraph_number": (
            metadata.get("paragraph_number")
            if metadata.get("paragraph_number") is not None
            else locator.get("paragraph_number")
        ),
        "book_unit_type": metadata.get(
            "book_unit_type"
        ),
        "heading_path": (
            metadata.get("heading_path")
            or locator.get("heading_path")
        ),
    }

def get_transcript_context(record):
    metadata = record.get("source_metadata")
    locator = record.get("locator")

    metadata = (
        metadata
        if isinstance(metadata, dict)
        else {}
    )

    locator = (
        locator
        if isinstance(locator, dict)
        else {}
    )

    return {
        "program_id": (
            metadata.get("program_id")
            or locator.get("program_id")
        ),
        "episode_number": locator.get(
            "episode_number"
        ),
        "utterance_index": metadata.get(
            "utterance_index"
        ),
        "record_type": metadata.get(
            "record_type"
        ),
    }

def normalized_token_sequence(text):
    return re.findall(
        r"[\w\u0600-\u06FF\u200c]+",
        text,
        flags=re.UNICODE,
    )

def structural_distance(
    record_a,
    record_b,
):
    source_type = record_a.get(
        "source_type"
    )

    if (
        source_type
        != record_b.get("source_type")
    ):
        return None

    if source_type == "book":
        a = get_book_context(record_a)
        b = get_book_context(record_b)

        if (
            a["edition_id"] is not None
            and a["edition_id"] == b["edition_id"]
            and isinstance(
                a["paragraph_number"],
                int,
            )
            and isinstance(
                b["paragraph_number"],
                int,
            )
        ):
            return abs(
                a["paragraph_number"]
                - b["paragraph_number"]
            )

    if source_type == "transcript":
        a = get_transcript_context(record_a)
        b = get_transcript_context(record_b)

        if (
            a["program_id"] is not None
            and a["program_id"] == b["program_id"]
            and a["episode_number"] is not None
            and a["episode_number"] == b["episode_number"]
            and isinstance(
                a["utterance_index"],
                int,
            )
            and isinstance(
                b["utterance_index"],
                int,
            )
        ):
            return abs(
                a["utterance_index"]
                - b["utterance_index"]
            )

    return None

def descriptive_relation(
    record_a,
    record_b,
    distance,
):
    source_type = record_a["source_type"]

    if (
        record_a["source_id"]
        == record_b["source_id"]
    ):
        if distance is not None:
            return "SAME_SOURCE_LOCAL"

        return "SAME_SOURCE_NONLOCAL"

    if (
        source_type == "transcript"
        and record_b["source_type"]
        == "transcript"
    ):
        a = get_transcript_context(record_a)
        b = get_transcript_context(record_b)

        if (
            a["program_id"] is not None
            and a["program_id"] == b["program_id"]
            and a["episode_number"]
            != b["episode_number"]
        ):
            return "SAME_PROGRAM_DIFFERENT_EPISODE"

    if (
        source_type == "book"
        and record_b["source_type"]
        == "book"
    ):
        a = get_book_context(record_a)
        b = get_book_context(record_b)

        if (
            a["work_id"] is not None
            and a["work_id"] == b["work_id"]
        ):
            return "SAME_WORK_DIFFERENT_SOURCE"

    return "DIFFERENT_SOURCE_ID"

# -------------------------------------------------------------------
# Context enrichment
# -------------------------------------------------------------------

context_enriched_pairs = []

for row in context_candidate_rows:
    record_a = context_record_lookup[
        row["document_id"]
    ]

    record_b = context_record_lookup[
        row["candidate_document_id"]
    ]

    text_a = row["text"].strip()
    text_b = row["candidate_text"].strip()

    tokens_a = normalized_token_sequence(
        text_a
    )
    tokens_b = normalized_token_sequence(
        text_b
    )

    token_set_a = set(tokens_a)
    token_set_b = set(tokens_b)

    distance = structural_distance(
        record_a,
        record_b,
    )

    relation = descriptive_relation(
        record_a,
        record_b,
        distance,
    )

    shorter, longer = sorted(
        [text_a, text_b],
        key=len,
    )

    exact_containment = (
        bool(shorter)
        and shorter in longer
        and shorter != longer
    )

    same_token_set = (
        bool(token_set_a or token_set_b)
        and token_set_a == token_set_b
    )

    no_informative_a = (
        row["informative_chars_a"] == 0
    )

    no_informative_b = (
        row["informative_chars_b"] == 0
    )

    context_enriched_pairs.append(
        {
            **row,
            "structural_relation": relation,
            "structural_distance": distance,
            "exact_containment": exact_containment,
            "same_token_set": same_token_set,
            "no_informative_a": no_informative_a,
            "no_informative_b": no_informative_b,
            "book_context_a": (
                get_book_context(record_a)
                if record_a["source_type"]
                == "book"
                else None
            ),
            "book_context_b": (
                get_book_context(record_b)
                if record_b["source_type"]
                == "book"
                else None
            ),
            "transcript_context_a": (
                get_transcript_context(
                    record_a
                )
                if record_a["source_type"]
                == "transcript"
                else None
            ),
            "transcript_context_b": (
                get_transcript_context(
                    record_b
                )
                if record_b["source_type"]
                == "transcript"
                else None
            ),
        }
    )

# -------------------------------------------------------------------
# Relation summary
# -------------------------------------------------------------------

print("=== STRUCTURAL CONTEXT PROFILE ===")
print(
    "Candidate pairs >= 0.90:",
    len(context_enriched_pairs),
)
print(
    "Documents required:",
    len(target_ids),
)
print(
    "Missing documents:",
    len(missing_context_ids),
)

relation_counts = Counter(
    row["structural_relation"]
    for row in context_enriched_pairs
)

print("\n=== STRUCTURAL RELATIONS ===")

for relation, count in sorted(
    relation_counts.items()
):
    print(
        f"- {relation}:",
        count,
    )

print("\n=== DIAGNOSTIC FLAGS ===")

diagnostic_counts = {
    "exact_containment": sum(
        row["exact_containment"]
        for row in context_enriched_pairs
    ),
    "same_token_set": sum(
        row["same_token_set"]
        for row in context_enriched_pairs
    ),
    "one_or_both_no_informative_chars": sum(
        row["no_informative_a"]
        or row["no_informative_b"]
        for row in context_enriched_pairs
    ),
    "structural_distance_available": sum(
        row["structural_distance"]
        is not None
        for row in context_enriched_pairs
    ),
}

for key, value in diagnostic_counts.items():
    print(
        f"- {key}:",
        value,
    )

# -------------------------------------------------------------------
# Structural-distance profile
# -------------------------------------------------------------------

print("\n=== SAME-SOURCE STRUCTURAL DISTANCES ===")

for source_type in [
    "book",
    "transcript",
]:
    distances = [
        row["structural_distance"]
        for row in context_enriched_pairs
        if row["source_type"] == source_type
        and row["structural_distance"]
        is not None
    ]

    print("\nSOURCE:", source_type)
    print(
        "Pairs with structural distance:",
        len(distances),
    )

    if distances:
        array = np.asarray(
            distances,
            dtype=float,
        )

        print(
            " Min:",
            int(np.min(array)),
        )
        print(
            " Median:",
            round(
                float(np.median(array)),
                2,
            ),
        )
        print(
            " P90:",
            round(
                float(
                    np.percentile(
                        array,
                        90,
                    )
                ),
                2,
            ),
        )
        print(
            " Max:",
            int(np.max(array)),
        )

# -------------------------------------------------------------------
# Representative inspection by relation
# -------------------------------------------------------------------

print("\n=== REPRESENTATIVE PAIRS BY STRUCTURAL RELATION ===")

relation_order = [
    "SAME_SOURCE_LOCAL",
    "SAME_SOURCE_NONLOCAL",
    "SAME_PROGRAM_DIFFERENT_EPISODE",
    "SAME_WORK_DIFFERENT_SOURCE",
    "DIFFERENT_SOURCE_ID",
]

for relation in relation_order:
    rows = [
        row
        for row in context_enriched_pairs
        if row["structural_relation"]
        == relation
    ]

    if not rows:
        continue

    rows.sort(
        key=lambda row: (
            -row["similarity"],
            -row["sequence_matcher"],
            row["document_id"],
        )
    )

    print("\n" + "=" * 100)
    print("RELATION:", relation)
    print("Pairs:", len(rows))

    for index, row in enumerate(
        rows[:12],
        start=1,
    ):
        print("\nPair", index)

        print(
            " Source:",
            row["source_type"],
        )

        print(
            " Character TF-IDF:",
            round(
                row["similarity"],
                6,
            ),
        )

        print(
            " SequenceMatcher:",
            round(
                row["sequence_matcher"],
                6,
            ),
        )

        print(
            " Token Jaccard:",
            round(
                row["token_jaccard"],
                6,
            ),
        )

        print(
            " Structural distance:",
            row["structural_distance"],
        )

        print(
            " Exact containment:",
            row["exact_containment"],
        )

        print(
            " Same token set:",
            row["same_token_set"],
        )

        print(
            " No-informative flags:",
            row["no_informative_a"],
            row["no_informative_b"],
        )

        print(
            " A:",
            row["document_id"],
        )

        print(
            " B:",
            row["candidate_document_id"],
        )

        print(
            " A text:",
            row["text"]
            .replace("\n", " ")[:500],
        )

        print(
            " B text:",
            row["candidate_text"]
            .replace("\n", " ")[:500],
        )

# -------------------------------------------------------------------
# Validation
# -------------------------------------------------------------------

pair_count_ok = (
    len(context_enriched_pairs)
    == len(context_candidate_rows)
)

all_relations_present = all(
    isinstance(
        row["structural_relation"],
        str,
    )
    and bool(
        row["structural_relation"]
    )
    for row in context_enriched_pairs
)

no_missing_records = (
    len(missing_context_ids) == 0
)

all_nonexact = all(
    row["text_sha256"]
    != row["candidate_text_sha256"]
    for row in context_enriched_pairs
)

print("\nValidation:")

print(
    "Context pair count preserved:",
    "PASS" if pair_count_ok else "FAIL",
)

print(
    "All required documents resolved:",
    "PASS" if no_missing_records else "FAIL",
)

print(
    "Structural relation assigned to every pair:",
    "PASS" if all_relations_present else "FAIL",
)

print(
    "Exact-text pairs remain excluded:",
    "PASS" if all_nonexact else "FAIL",
)

cell_gate = (
    pair_count_ok
    and no_missing_records
    and all_relations_present
    and all_nonexact
)

print(
    "CODE CELL 16 GATE:",
    "PASS" if cell_gate else "FAIL",
)
```

<div style="direction: rtl; white-space: normal; line-height: 1.7;">

<h3>تحلیل Connected Componentهای Candidateهای Near-duplicate</h3>

<p>
تا این مرحله Candidateهای Near-duplicate عمدتاً به‌صورت Pair بررسی شده‌اند.
اما Pairwise inspection به‌تنهایی نشان نمی‌دهد که یک شباهت فقط میان دو Document رخ داده یا بخشی از یک خانواده بزرگ‌تر از متن‌های مشابه است.
</p>

<p>
برای مثال Boilerplate یک برنامه رادیویی ممکن است در تعداد زیادی Episode ظاهر شود.
در این حالت ده‌ها Pair جداگانه در واقع متعلق به یک Family واحد هستند.
همین مسئله می‌تواند در Book برای Worksheet templateها یا در Website برای مجموعه مقاله‌های دارای قالب مشترک رخ دهد.
</p>

<p>
در این مرحله Candidateهای دارای Character TF-IDF حداقل <code>0.90</code> به یک Graph بدون جهت تبدیل می‌شوند:
هر Document یک Node و هر Candidate Pair یک Edge است.
سپس Connected Componentهای Graph استخراج می‌شوند.
</p>

<p>
این مقدار <code>0.90</code> همچنان فقط همان Boundary آزمایشی Profiling مراحل قبل است و Threshold تولید یا Rule حذف محسوب نمی‌شود.
</p>

<h4>هدف تحلیل Component</h4>

<ul>
<li>تشخیص اینکه Candidateها منفرد هستند یا Familyهای بزرگ تشکیل می‌دهند.</li>
<li>اندازه‌گیری تعداد Document و Source ID در هر Component.</li>
<li>بررسی اینکه Component در یک Source باقی مانده یا چند Source ID را متصل کرده است.</li>
<li>بررسی توزیع Similarity درون هر Component.</li>
<li>مشاهده تعداد Relationهای ساختاری درون هر Component.</li>
<li>مشاهده Low-information، Containment و Same-token-set در سطح Family.</li>
<li>شناسایی خانواده‌های محتمل Boilerplate، Reuse یا Template برای Inspection بعدی، بدون Classification خودکار.</li>
</ul>

<h4>اصل مهم</h4>

<p>
Connected Component به معنی Duplicate قطعی نیست.
اگر A به B شبیه باشد و B به C شبیه باشد، قرار گرفتن A، B و C در یک Component فقط وجود یک زنجیره شباهت را اثبات می‌کند.
این مرحله هیچ Transitive duplicate relation ایجاد نمی‌کند.
</p>

<h4>معیار PASS</h4>

<p>
تمام Edgeهای Candidateهای مرحله قبل باید دقیقاً در Graph حضور داشته باشند،
تمام Nodeها باید به یک Component تخصیص داده شوند،
هیچ Document خارج از Candidate population نباید وارد Graph شود
و Component accounting باید کامل و deterministic باشد.
</p>

</div>

```python
# Code Cell 17

from collections import Counter, defaultdict
import math

# -------------------------------------------------------------------
# Candidate graph
# Reuses the profiling-only >= 0.90 population from Code Cell 16.
# -------------------------------------------------------------------

graph_edges = context_enriched_pairs

graph_nodes = {
    document_id
    for row in graph_edges
    for document_id in [
        row["document_id"],
        row["candidate_document_id"],
    ]
}

# -------------------------------------------------------------------
# Deterministic union-find
# -------------------------------------------------------------------

parent = {
    node: node
    for node in graph_nodes
}

rank = {
    node: 0
    for node in graph_nodes
}

def find(node):
    if parent[node] != node:
        parent[node] = find(parent[node])

    return parent[node]

def union(node_a, node_b):
    root_a = find(node_a)
    root_b = find(node_b)

    if root_a == root_b:
        return

    if rank[root_a] < rank[root_b]:
        root_a, root_b = root_b, root_a

    parent[root_b] = root_a

    if rank[root_a] == rank[root_b]:
        rank[root_a] += 1

for row in graph_edges:
    union(
        row["document_id"],
        row["candidate_document_id"],
    )

# -------------------------------------------------------------------
# Build components
# -------------------------------------------------------------------

component_members = defaultdict(list)

for node in sorted(graph_nodes):
    component_members[
        find(node)
    ].append(node)

components = list(
    component_members.values()
)

components.sort(
    key=lambda members: (
        -len(members),
        members[0],
    )
)

node_to_component = {}

for component_index, members in enumerate(
    components,
    start=1,
):
    for document_id in members:
        node_to_component[
            document_id
        ] = component_index

component_edges = defaultdict(list)

for row in graph_edges:
    component_id_a = node_to_component[
        row["document_id"]
    ]

    component_id_b = node_to_component[
        row["candidate_document_id"]
    ]

    if component_id_a != component_id_b:
        raise AssertionError(
            "Graph edge spans two different components."
        )

    component_edges[
        component_id_a
    ].append(row)

# -------------------------------------------------------------------
# Component profiles
# -------------------------------------------------------------------

component_profiles = []

for component_index, members in enumerate(
    components,
    start=1,
):
    edges = component_edges[
        component_index
    ]

    records = [
        context_record_lookup[
            document_id
        ]
        for document_id in members
    ]

    source_types = sorted({
        record["source_type"]
        for record in records
    })

    source_ids = sorted({
        record["source_id"]
        for record in records
    })

    titles = sorted({
        record.get("title")
        for record in records
        if record.get("title") is not None
    })

    relation_counts = Counter(
        edge["structural_relation"]
        for edge in edges
    )

    similarities = [
        edge["similarity"]
        for edge in edges
    ]

    sequence_scores = [
        edge["sequence_matcher"]
        for edge in edges
    ]

    low_information_edges = sum(
        edge["no_informative_a"]
        or edge["no_informative_b"]
        for edge in edges
    )

    containment_edges = sum(
        edge["exact_containment"]
        for edge in edges
    )

    same_token_set_edges = sum(
        edge["same_token_set"]
        for edge in edges
    )

    transcript_program_ids = set()
    book_work_ids = set()

    for record in records:
        if record["source_type"] == "transcript":
            context = get_transcript_context(
                record
            )

            if context["program_id"] is not None:
                transcript_program_ids.add(
                    context["program_id"]
                )

        if record["source_type"] == "book":
            context = get_book_context(
                record
            )

            if context["work_id"] is not None:
                book_work_ids.add(
                    context["work_id"]
                )

    component_profiles.append(
        {
            "component_id": component_index,
            "member_count": len(members),
            "edge_count": len(edges),
            "members": members,
            "source_types": source_types,
            "source_ids": source_ids,
            "source_id_count": len(source_ids),
            "titles": titles,
            "relation_counts": dict(
                sorted(
                    relation_counts.items()
                )
            ),
            "similarity_min": min(
                similarities
            ),
            "similarity_median": float(
                np.median(
                    similarities
                )
            ),
            "similarity_max": max(
                similarities
            ),
            "sequence_min": min(
                sequence_scores
            ),
            "sequence_median": float(
                np.median(
                    sequence_scores
                )
            ),
            "sequence_max": max(
                sequence_scores
            ),
            "low_information_edges": (
                low_information_edges
            ),
            "containment_edges": (
                containment_edges
            ),
            "same_token_set_edges": (
                same_token_set_edges
            ),
            "transcript_program_ids": sorted(
                transcript_program_ids
            ),
            "book_work_ids": sorted(
                book_work_ids
            ),
        }
    )

# -------------------------------------------------------------------
# Global graph summary
# -------------------------------------------------------------------

print("=== NEAR-DUPLICATE CANDIDATE GRAPH ===")
print(
    "Graph nodes:",
    len(graph_nodes),
)
print(
    "Graph edges:",
    len(graph_edges),
)
print(
    "Connected components:",
    len(component_profiles),
)

component_size_counts = Counter(
    profile["member_count"]
    for profile in component_profiles
)

print("\n=== COMPONENT SIZE DISTRIBUTION ===")

for size, count in sorted(
    component_size_counts.items()
):
    print(
        f"- size {size}:",
        count,
    )

print("\n=== LARGEST COMPONENTS ===")

for profile in component_profiles[:25]:
    print("\n" + "=" * 100)

    print(
        "COMPONENT:",
        profile["component_id"],
    )

    print(
        "Members:",
        profile["member_count"],
    )

    print(
        "Edges:",
        profile["edge_count"],
    )

    print(
        "Source types:",
        profile["source_types"],
    )

    print(
        "Unique source IDs:",
        profile["source_id_count"],
    )

    print(
        "Transcript programs:",
        profile["transcript_program_ids"],
    )

    print(
        "Book works:",
        profile["book_work_ids"],
    )

    print(
        "Relation counts:",
        profile["relation_counts"],
    )

    print(
        "Similarity min/median/max:",
        round(
            profile["similarity_min"],
            6,
        ),
        "/",
        round(
            profile["similarity_median"],
            6,
        ),
        "/",
        round(
            profile["similarity_max"],
            6,
        ),
    )

    print(
        "Sequence min/median/max:",
        round(
            profile["sequence_min"],
            6,
        ),
        "/",
        round(
            profile["sequence_median"],
            6,
        ),
        "/",
        round(
            profile["sequence_max"],
            6,
        ),
    )

    print(
        "Low-information edges:",
        profile["low_information_edges"],
    )

    print(
        "Containment edges:",
        profile["containment_edges"],
    )

    print(
        "Same-token-set edges:",
        profile["same_token_set_edges"],
    )

    print("\nMembers:")

    for document_id in profile[
        "members"
    ][:30]:
        record = context_record_lookup[
            document_id
        ]

        text = (
            record.get("text")
            or ""
        ).replace(
            "\n",
            " ",
        ).strip()

        print(
            "-",
            document_id,
            "| source_id:",
            record.get("source_id"),
            "| text:",
            text[:220],
        )

    if profile["member_count"] > 30:
        print(
            "... remaining members:",
            profile["member_count"] - 30,
        )

    print("\nRepresentative edges:")

    edges = sorted(
        component_edges[
            profile["component_id"]
        ],
        key=lambda row: (
            -row["similarity"],
            -row["sequence_matcher"],
            row["document_id"],
        ),
    )

    for edge in edges[:12]:
        print(
            "-",
            edge["document_id"],
            "<->",
            edge["candidate_document_id"],
        )

        print(
            "  char:",
            round(
                edge["similarity"],
                6,
            ),
            "| seq:",
            round(
                edge["sequence_matcher"],
                6,
            ),
            "| relation:",
            edge["structural_relation"],
            "| containment:",
            edge["exact_containment"],
            "| same tokens:",
            edge["same_token_set"],
        )

# -------------------------------------------------------------------
# Component source-pattern summary
# -------------------------------------------------------------------

print("\n=== COMPONENT SOURCE PATTERNS ===")

source_pattern_counts = Counter()

for profile in component_profiles:
    if len(
        profile["source_types"]
    ) > 1:
        pattern = "MULTI_SOURCE_TYPE"

    elif profile["source_id_count"] == 1:
        pattern = "ONE_SOURCE_ID"

    else:
        pattern = "MULTIPLE_SOURCE_IDS"

    source_pattern_counts[
        pattern
    ] += 1

for pattern, count in sorted(
    source_pattern_counts.items()
):
    print(
        f"- {pattern}:",
        count,
    )

# -------------------------------------------------------------------
# Validation
# -------------------------------------------------------------------

all_edge_nodes_present = all(
    row["document_id"] in graph_nodes
    and row["candidate_document_id"]
    in graph_nodes
    for row in graph_edges
)

all_nodes_assigned = (
    len(node_to_component)
    == len(graph_nodes)
)

component_node_accounting_ok = (
    sum(
        profile["member_count"]
        for profile in component_profiles
    )
    == len(graph_nodes)
)

component_edge_accounting_ok = (
    sum(
        profile["edge_count"]
        for profile in component_profiles
    )
    == len(graph_edges)
)

all_profiles_finite = all(
    math.isfinite(
        profile[key]
    )
    for profile in component_profiles
    for key in [
        "similarity_min",
        "similarity_median",
        "similarity_max",
        "sequence_min",
        "sequence_median",
        "sequence_max",
    ]
)

print("\nValidation:")

print(
    "All edge nodes present:",
    "PASS" if all_edge_nodes_present else "FAIL",
)

print(
    "All graph nodes assigned:",
    "PASS" if all_nodes_assigned else "FAIL",
)

print(
    "Component node accounting:",
    "PASS" if component_node_accounting_ok else "FAIL",
)

print(
    "Component edge accounting:",
    "PASS" if component_edge_accounting_ok else "FAIL",
)

print(
    "All component metrics finite:",
    "PASS" if all_profiles_finite else "FAIL",
)

cell_gate = (
    all_edge_nodes_present
    and all_nodes_assigned
    and component_node_accounting_ok
    and component_edge_accounting_ok
    and all_profiles_finite
)

print(
    "CODE CELL 17 GATE:",
    "PASS" if cell_gate else "FAIL",
)
```

<h3>تحلیل Position و الگوی تکرار Componentها در Source</h3>

<p>
Connected Componentها نشان دادند که بخشی از Candidateهای مشابه به‌صورت Familyهای چندعضوی تکرار می‌شوند.
اما برای تشخیص ماهیت این Familyها باید بدانیم هر Member در کجای Source خود قرار گرفته است.
</p>

<p>
این موضوع برای Transcript اهمیت ویژه دارد.
اگر یک Family مشابه در Episodeهای متعدد یک Program ظاهر شود و اعضای آن تقریباً همیشه در ابتدای Episode یا انتهای Episode باشند، این یک Evidence قوی برای Program Boilerplate است.
</p>

<p>
در مقابل، اگر عبارت‌های کوتاه مشابه در نقاط پراکنده Episodeها دیده شوند، صرف تکرار آن‌ها برای Boilerplate classification کافی نیست.
</p>

<h4>روش Position Profiling</h4>

<ul>
<li>برای هر <code>source_id</code> تعداد Documentهای واقعی از Dataset محاسبه می‌شود.</li>
<li>برای Transcript از <code>utterance_index</code> و برای Book از <code>paragraph_number</code> در صورت موجود بودن استفاده می‌شود.</li>
<li>Position هر Document نسبت به محدوده واقعی همان Source به مقدار <code>0..1</code> تبدیل می‌شود.</li>
<li>برای Profiling توصیفی، Position به سه Bucket تقسیم می‌شود: ۱۰٪ ابتدایی، بخش میانی و ۱۰٪ انتهایی.</li>
<li>این ۱۰٪ صرفاً Bucket توصیفی برای Discovery است و Rule تولید یا Threshold حذف نیست.</li>
</ul>

<h4>Evidenceهای Component-level</h4>

<ul>
<li>تعداد Source IDهای مختلف در Component.</li>
<li>تعداد Programهای Transcript.</li>
<li>توزیع Memberها در START / MIDDLE / END.</li>
<li>میزان تمرکز Position اعضای Component.</li>
<li>تعداد Memberهای Component که در Sourceهای متفاوت تکرار شده‌اند.</li>
</ul>

<h4>هدف</h4>

<p>
هدف این مرحله جداکردن الگوهایی مانند
Program Boilerplate،
عبارت‌های عمومی پراکنده،
تکرار داخل یک Episode،
و Templateهای Book
از یکدیگر است.
هنوز هیچ Component به‌صورت خودکار حذف، Merge یا Quarantine نمی‌شود.
</p>

<h4>معیار PASS</h4>

<p>
تمام Memberهایی که Position ساختاری قابل محاسبه دارند باید Position معتبر در محدوده <code>0..1</code> دریافت کنند،
Component accounting باید حفظ شود
و هیچ Relation یا Document از مرحله قبل تغییر نکند.
</p>

</div>

```python
# Code Cell 18

from collections import Counter, defaultdict
import json
import math

# -------------------------------------------------------------------
# Rebuild complete structural position ranges for every source_id
# -------------------------------------------------------------------

source_position_values = defaultdict(list)
source_document_counts = Counter()

all_position_records = {}

with input_path.open("r", encoding="utf-8") as handle:
    for line in handle:
        record = json.loads(line)

        document_id = record["document_id"]
        source_id = record["source_id"]
        source_type = record["source_type"]

        source_document_counts[source_id] += 1

        position_value = None

        if source_type == "transcript":
            context = get_transcript_context(
                record
            )

            if isinstance(
                context["utterance_index"],
                int,
            ):
                position_value = context[
                    "utterance_index"
                ]

        elif source_type == "book":
            context = get_book_context(
                record
            )

            if isinstance(
                context["paragraph_number"],
                int,
            ):
                position_value = context[
                    "paragraph_number"
                ]

        if position_value is not None:
            source_position_values[
                source_id
            ].append(
                position_value
            )

        all_position_records[
            document_id
        ] = {
            "source_type": source_type,
            "source_id": source_id,
            "raw_position": position_value,
        }

# -------------------------------------------------------------------
# Source-level min/max ranges
# -------------------------------------------------------------------

source_position_ranges = {}

for source_id, values in source_position_values.items():
    source_position_ranges[
        source_id
    ] = {
        "min": min(values),
        "max": max(values),
        "count_with_position": len(values),
        "document_count": source_document_counts[
            source_id
        ],
    }

def relative_position(
    source_id,
    raw_position,
):
    if raw_position is None:
        return None

    position_range = source_position_ranges.get(
        source_id
    )

    if position_range is None:
        return None

    minimum = position_range["min"]
    maximum = position_range["max"]

    if maximum == minimum:
        return 0.0

    return (
        raw_position - minimum
    ) / (
        maximum - minimum
    )

def position_bucket(value):
    if value is None:
        return "UNKNOWN"

    # Descriptive profiling buckets only.
    if value <= 0.10:
        return "START_10_PERCENT"

    if value >= 0.90:
        return "END_10_PERCENT"

    return "MIDDLE_80_PERCENT"

# -------------------------------------------------------------------
# Enrich component members with position evidence
# -------------------------------------------------------------------

component_position_profiles = []

for profile in component_profiles:
    member_positions = []

    for document_id in profile["members"]:
        base = all_position_records[
            document_id
        ]

        raw_position = base[
            "raw_position"
        ]

        relative = relative_position(
            base["source_id"],
            raw_position,
        )

        member_positions.append(
            {
                "document_id": document_id,
                "source_type": base[
                    "source_type"
                ],
                "source_id": base[
                    "source_id"
                ],
                "raw_position": raw_position,
                "relative_position": relative,
                "position_bucket": position_bucket(
                    relative
                ),
            }
        )

    known_positions = [
        item["relative_position"]
        for item in member_positions
        if item["relative_position"]
        is not None
    ]

    bucket_counts = Counter(
        item["position_bucket"]
        for item in member_positions
    )

    if known_positions:
        position_min = min(
            known_positions
        )
        position_max = max(
            known_positions
        )
        position_median = float(
            np.median(
                known_positions
            )
        )
        position_span = (
            position_max
            - position_min
        )
    else:
        position_min = None
        position_max = None
        position_median = None
        position_span = None

    component_position_profiles.append(
        {
            **profile,
            "member_positions": (
                member_positions
            ),
            "known_position_count": len(
                known_positions
            ),
            "position_bucket_counts": dict(
                sorted(
                    bucket_counts.items()
                )
            ),
            "position_min": position_min,
            "position_median": position_median,
            "position_max": position_max,
            "position_span": position_span,
        }
    )

# -------------------------------------------------------------------
# Global position coverage
# -------------------------------------------------------------------

print("=== STRUCTURAL POSITION COVERAGE ===")

for source_type in [
    "website",
    "book",
    "transcript",
]:
    relevant_members = [
        item
        for profile
        in component_position_profiles
        for item in profile[
            "member_positions"
        ]
        if item["source_type"]
        == source_type
    ]

    known = [
        item
        for item in relevant_members
        if item["relative_position"]
        is not None
    ]

    print("\nSOURCE:", source_type)
    print(
        "Component members:",
        len(relevant_members),
    )
    print(
        "Members with structural position:",
        len(known),
    )

    if relevant_members:
        print(
            "Coverage:",
            round(
                len(known)
                / len(relevant_members),
                6,
            ),
        )

# -------------------------------------------------------------------
# Transcript multi-source family profiling
# -------------------------------------------------------------------

transcript_family_profiles = [
    profile
    for profile in component_position_profiles
    if profile["source_types"]
    == ["transcript"]
    and profile["source_id_count"] >= 2
]

transcript_family_profiles.sort(
    key=lambda profile: (
        -profile["member_count"],
        profile["component_id"],
    )
)

print(
    "\n=== TRANSCRIPT MULTI-SOURCE COMPONENT POSITION PROFILE ==="
)
print(
    "Components:",
    len(transcript_family_profiles),
)

for profile in transcript_family_profiles[:30]:
    print("\n" + "=" * 100)

    print(
        "COMPONENT:",
        profile["component_id"],
    )

    print(
        "Members:",
        profile["member_count"],
    )

    print(
        "Source IDs:",
        profile["source_id_count"],
    )

    print(
        "Programs:",
        profile[
            "transcript_program_ids"
        ],
    )

    print(
        "Relation counts:",
        profile["relation_counts"],
    )

    print(
        "Position buckets:",
        profile[
            "position_bucket_counts"
        ],
    )

    print(
        "Known positions:",
        profile[
            "known_position_count"
        ],
    )

    print(
        "Relative position min/median/max:",
        (
            round(
                profile["position_min"],
                6,
            )
            if profile[
                "position_min"
            ] is not None
            else None
        ),
        "/",
        (
            round(
                profile[
                    "position_median"
                ],
                6,
            )
            if profile[
                "position_median"
            ] is not None
            else None
        ),
        "/",
        (
            round(
                profile["position_max"],
                6,
            )
            if profile[
                "position_max"
            ] is not None
            else None
        ),
    )

    print(
        "Relative position span:",
        (
            round(
                profile["position_span"],
                6,
            )
            if profile[
                "position_span"
            ] is not None
            else None
        ),
    )

    print("\nMembers:")

    for item in profile[
        "member_positions"
    ][:20]:
        record = context_record_lookup[
            item["document_id"]
        ]

        text = (
            record.get("text")
            or ""
        ).replace(
            "\n",
            " ",
        ).strip()

        print(
            "-",
            item["document_id"],
        )

        print(
            "  source_id:",
            item["source_id"],
        )

        print(
            "  raw position:",
            item["raw_position"],
        )

        print(
            "  relative:",
            (
                round(
                    item[
                        "relative_position"
                    ],
                    6,
                )
                if item[
                    "relative_position"
                ] is not None
                else None
            ),
        )

        print(
            "  bucket:",
            item["position_bucket"],
        )

        print(
            "  text:",
            text[:220],
        )

# -------------------------------------------------------------------
# Book component positional profile
# -------------------------------------------------------------------

book_family_profiles = [
    profile
    for profile in component_position_profiles
    if profile["source_types"]
    == ["book"]
]

book_family_profiles.sort(
    key=lambda profile: (
        -profile["member_count"],
        profile["component_id"],
    )
)

print(
    "\n=== BOOK COMPONENT POSITION PROFILE ==="
)

for profile in book_family_profiles[:20]:
    print("\n" + "=" * 100)

    print(
        "COMPONENT:",
        profile["component_id"],
    )

    print(
        "Members:",
        profile["member_count"],
    )

    print(
        "Source IDs:",
        profile["source_id_count"],
    )

    print(
        "Works:",
        profile["book_work_ids"],
    )

    print(
        "Position buckets:",
        profile[
            "position_bucket_counts"
        ],
    )

    print(
        "Relative position min/median/max:",
        (
            round(
                profile["position_min"],
                6,
            )
            if profile[
                "position_min"
            ] is not None
            else None
        ),
        "/",
        (
            round(
                profile[
                    "position_median"
                ],
                6,
            )
            if profile[
                "position_median"
            ] is not None
            else None
        ),
        "/",
        (
            round(
                profile["position_max"],
                6,
            )
            if profile[
                "position_max"
            ] is not None
            else None
        ),
    )

# -------------------------------------------------------------------
# Descriptive positional-pattern counts
# -------------------------------------------------------------------

print(
    "\n=== DESCRIPTIVE COMPONENT POSITION PATTERNS ==="
)

position_pattern_counts = Counter()

for profile in component_position_profiles:
    buckets = profile[
        "position_bucket_counts"
    ]

    known = profile[
        "known_position_count"
    ]

    if known == 0:
        pattern = "NO_POSITION"

    elif (
        buckets.get(
            "START_10_PERCENT",
            0,
        )
        == known
    ):
        pattern = "ALL_START"

    elif (
        buckets.get(
            "END_10_PERCENT",
            0,
        )
        == known
    ):
        pattern = "ALL_END"

    elif (
        buckets.get(
            "MIDDLE_80_PERCENT",
            0,
        )
        == known
    ):
        pattern = "ALL_MIDDLE"

    else:
        pattern = "MIXED_POSITION"

    position_pattern_counts[
        pattern
    ] += 1

for pattern, count in sorted(
    position_pattern_counts.items()
):
    print(
        f"- {pattern}:",
        count,
    )

# -------------------------------------------------------------------
# Validation
# -------------------------------------------------------------------

position_values = [
    item["relative_position"]
    for profile in component_position_profiles
    for item in profile[
        "member_positions"
    ]
    if item["relative_position"]
    is not None
]

all_positions_finite = all(
    math.isfinite(value)
    for value in position_values
)

all_positions_in_range = all(
    -1e-12
    <= value
    <= 1.0 + 1e-12
    for value in position_values
)

component_count_preserved = (
    len(component_position_profiles)
    == len(component_profiles)
)

member_accounting_preserved = (
    sum(
        profile["member_count"]
        for profile
        in component_position_profiles
    )
    == len(graph_nodes)
)

component_ids_preserved = (
    {
        profile["component_id"]
        for profile
        in component_position_profiles
    }
    == {
        profile["component_id"]
        for profile
        in component_profiles
    }
)

print("\nValidation:")

print(
    "Component count preserved:",
    "PASS"
    if component_count_preserved
    else "FAIL",
)

print(
    "Member accounting preserved:",
    "PASS"
    if member_accounting_preserved
    else "FAIL",
)

print(
    "Component IDs preserved:",
    "PASS"
    if component_ids_preserved
    else "FAIL",
)

print(
    "All structural positions finite:",
    "PASS"
    if all_positions_finite
    else "FAIL",
)

print(
    "All structural positions within [0, 1]:",
    "PASS"
    if all_positions_in_range
    else "FAIL",
)

cell_gate = (
    component_count_preserved
    and member_accounting_preserved
    and component_ids_preserved
    and all_positions_finite
    and all_positions_in_range
)

print(
    "CODE CELL 18 GATE:",
    "PASS" if cell_gate else "FAIL",
)
```

<h3>اعتبارسنجی مستقل Text Hash و Exact-duplicate Population</h3>

<p>
تا این مرحله Exact Duplicateها با استفاده از فیلد
<code>text_sha256</code>
شناسایی شده‌اند.
پیش از تبدیل نتایج Discovery به Policy یا Artifact نهایی باید به‌صورت مستقل اثبات شود که این Hashها واقعاً از متن فعلی هر Document تولید شده‌اند.
</p>

<p>
در این مرحله مقدار SHA-256 تمام <code>19,456</code> Document مستقیماً از
<code>text</code>
محاسبه و با مقدار ذخیره‌شده در
<code>text_sha256</code>
مقایسه می‌شود.
</p>

<p>
سپس Exact Duplicate population نیز از Hashهای تازه محاسبه‌شده بازسازی می‌شود و با نتایج Discovery قبلی مقایسه می‌شود.
این مرحله هیچ داده‌ای را تغییر نمی‌دهد.
</p>

<h4>بررسی‌های مستقل</h4>

<ul>
<li>محاسبه مجدد SHA-256 برای تمام متن‌ها.</li>
<li>شمارش Hash mismatchها.</li>
<li>بررسی اینکه یک Hash واحد هرگز به دو متن متفاوت اشاره نکند.</li>
<li>بازسازی تعداد Unique text hashها.</li>
<li>بازسازی تعداد Exact duplicate groupها.</li>
<li>بازسازی تعداد اعضای Exact duplicate groupها.</li>
<li>بازسازی توزیع اعضای Duplicate به تفکیک Website، Book و Transcript.</li>
<li>بازسازی Relation گروه‌ها بر اساس Source Type و Source ID.</li>
</ul>

<h4>چرا این Gate ضروری است؟</h4>

<p>
اگر حتی بخشی از <code>text_sha256</code>ها با متن فعلی ناسازگار باشند،
تمام Exact-deduplication analysis باید متوقف و اصلاح شود.
بنابراین هیچ Policy حذف، Canonical relation یا Accepted Dataset نباید قبل از این Revalidation ساخته شود.
</p>

<h4>معیار PASS</h4>

<p>
تمام Hashهای ذخیره‌شده باید با SHA-256 مستقل متن برابر باشند،
هیچ Hash collision-like inconsistency نباید مشاهده شود
و Population مستقل Exact Duplicate باید با Discovery قبلی سازگار باشد:
<code>192</code> گروه،
<code>931</code> عضو،
و <code>18,717</code> متن یکتای Hash‌شده.
</p>

</div>

```python
# Code Cell 19

from collections import Counter, defaultdict
from pathlib import Path
import hashlib
import json
import subprocess

def find_repo_root():
    result = subprocess.run(
        [
            "git",
            "rev-parse",
            "--show-toplevel",
        ],
        capture_output=True,
        text=True,
        check=True,
    )

    return Path(
        result.stdout.strip()
    ).resolve()

repo_root = find_repo_root()

hash_input_path = (
    repo_root
    / "data/phase2/processed/documents_normalized_v1.jsonl"
)

# -------------------------------------------------------------------
# Independent SHA-256 recomputation
# -------------------------------------------------------------------

hash_validation_records = []

stored_hash_mismatches = []

recomputed_hash_to_texts = defaultdict(set)

with hash_input_path.open(
    "r",
    encoding="utf-8",
) as handle:
    for line_number, line in enumerate(
        handle,
        start=1,
    ):
        record = json.loads(line)

        text = record["text"]

        recomputed_hash = hashlib.sha256(
            text.encode("utf-8")
        ).hexdigest()

        stored_hash = record[
            "text_sha256"
        ]

        if recomputed_hash != stored_hash:
            stored_hash_mismatches.append(
                {
                    "line_number": line_number,
                    "document_id": record[
                        "document_id"
                    ],
                    "stored_hash": stored_hash,
                    "recomputed_hash": (
                        recomputed_hash
                    ),
                }
            )

        recomputed_hash_to_texts[
            recomputed_hash
        ].add(text)

        hash_validation_records.append(
            {
                "document_id": record[
                    "document_id"
                ],
                "source_type": record[
                    "source_type"
                ],
                "source_id": record[
                    "source_id"
                ],
                "stored_hash": stored_hash,
                "recomputed_hash": (
                    recomputed_hash
                ),
                "text": text,
            }
        )

# -------------------------------------------------------------------
# Collision-like consistency check
# -------------------------------------------------------------------

hash_text_inconsistencies = {
    text_hash: texts
    for text_hash, texts
    in recomputed_hash_to_texts.items()
    if len(texts) > 1
}

# -------------------------------------------------------------------
# Independent exact-duplicate grouping
# -------------------------------------------------------------------

recomputed_groups = defaultdict(list)

for record in hash_validation_records:
    recomputed_groups[
        record["recomputed_hash"]
    ].append(record)

exact_duplicate_groups = {
    text_hash: members
    for text_hash, members
    in recomputed_groups.items()
    if len(members) > 1
}

exact_duplicate_member_count = sum(
    len(members)
    for members
    in exact_duplicate_groups.values()
)

unique_text_hash_count = len(
    recomputed_groups
)

duplicate_members_by_source = Counter()

relation_counts = Counter()

for members in exact_duplicate_groups.values():
    for member in members:
        duplicate_members_by_source[
            member["source_type"]
        ] += 1

    source_types = {
        member["source_type"]
        for member in members
    }

    source_ids = {
        member["source_id"]
        for member in members
    }

    if len(source_types) > 1:
        relation = "CROSS_SOURCE_TYPE"

    elif len(source_ids) > 1:
        relation = (
            "SAME_SOURCE_TYPE_DIFFERENT_SOURCE_ID"
        )

    else:
        relation = "SAME_SOURCE_ID"

    relation_counts[
        relation
    ] += 1

# -------------------------------------------------------------------
# Stored-hash grouping for direct equality comparison
# -------------------------------------------------------------------

stored_groups = defaultdict(list)

for record in hash_validation_records:
    stored_groups[
        record["stored_hash"]
    ].append(
        record["document_id"]
    )

recomputed_group_ids = {
    text_hash: sorted(
        member["document_id"]
        for member in members
    )
    for text_hash, members
    in recomputed_groups.items()
}

stored_group_ids = {
    text_hash: sorted(
        document_ids
    )
    for text_hash, document_ids
    in stored_groups.items()
}

grouping_identity_ok = (
    stored_group_ids
    == recomputed_group_ids
)

# -------------------------------------------------------------------
# Report
# -------------------------------------------------------------------

print("=== INDEPENDENT TEXT HASH REVALIDATION ===")
print(
    "Input:",
    hash_input_path.relative_to(
        repo_root
    ),
)
print(
    "Documents:",
    len(hash_validation_records),
)
print(
    "Stored/recomputed hash mismatches:",
    len(stored_hash_mismatches),
)

if stored_hash_mismatches:
    print("\nFirst mismatches:")

    for mismatch in stored_hash_mismatches[
        :20
    ]:
        print(mismatch)

print(
    "Collision-like hash/text inconsistencies:",
    len(hash_text_inconsistencies),
)

if hash_text_inconsistencies:
    print(
        "\nUnexpected hashes mapping to multiple texts:"
    )

    for text_hash, texts in list(
        hash_text_inconsistencies.items()
    )[:10]:
        print(
            text_hash,
            "| distinct texts:",
            len(texts),
        )

print("\n=== INDEPENDENT EXACT-DUPLICATE POPULATION ===")

print(
    "Unique recomputed text hashes:",
    unique_text_hash_count,
)

print(
    "Exact duplicate groups:",
    len(exact_duplicate_groups),
)

print(
    "Exact duplicate members:",
    exact_duplicate_member_count,
)

print("\nDuplicate members by source:")

for source_type in [
    "website",
    "book",
    "transcript",
]:
    print(
        f"- {source_type}:",
        duplicate_members_by_source.get(
            source_type,
            0,
        ),
    )

print("\nExact-group relations:")

for relation in [
    "SAME_SOURCE_ID",
    "SAME_SOURCE_TYPE_DIFFERENT_SOURCE_ID",
    "CROSS_SOURCE_TYPE",
]:
    print(
        f"- {relation}:",
        relation_counts.get(
            relation,
            0,
        ),
    )

# -------------------------------------------------------------------
# Independent expected-value gates from prior Discovery
# -------------------------------------------------------------------

record_count_ok = (
    len(hash_validation_records)
    == 19456
)

hash_integrity_ok = (
    len(stored_hash_mismatches)
    == 0
)

collision_consistency_ok = (
    len(hash_text_inconsistencies)
    == 0
)

unique_hash_count_ok = (
    unique_text_hash_count
    == 18717
)

duplicate_group_count_ok = (
    len(exact_duplicate_groups)
    == 192
)

duplicate_member_count_ok = (
    exact_duplicate_member_count
    == 931
)

source_member_counts_ok = (
    duplicate_members_by_source[
        "website"
    ]
    == 2
    and duplicate_members_by_source[
        "book"
    ]
    == 577
    and duplicate_members_by_source[
        "transcript"
    ]
    == 352
)

relation_counts_ok = (
    relation_counts[
        "SAME_SOURCE_ID"
    ]
    == 91
    and relation_counts[
        "SAME_SOURCE_TYPE_DIFFERENT_SOURCE_ID"
    ]
    == 92
    and relation_counts[
        "CROSS_SOURCE_TYPE"
    ]
    == 9
)

print("\nValidation:")

print(
    "Document count:",
    "PASS"
    if record_count_ok
    else "FAIL",
)

print(
    "Stored text_sha256 matches independently recomputed SHA-256:",
    "PASS"
    if hash_integrity_ok
    else "FAIL",
)

print(
    "One recomputed hash maps to one exact text:",
    "PASS"
    if collision_consistency_ok
    else "FAIL",
)

print(
    "Stored/recomputed grouping identity:",
    "PASS"
    if grouping_identity_ok
    else "FAIL",
)

print(
    "Unique hash population:",
    "PASS"
    if unique_hash_count_ok
    else "FAIL",
)

print(
    "Exact duplicate group count:",
    "PASS"
    if duplicate_group_count_ok
    else "FAIL",
)

print(
    "Exact duplicate member count:",
    "PASS"
    if duplicate_member_count_ok
    else "FAIL",
)

print(
    "Duplicate source distribution:",
    "PASS"
    if source_member_counts_ok
    else "FAIL",
)

print(
    "Exact-group relation distribution:",
    "PASS"
    if relation_counts_ok
    else "FAIL",
)

cell_gate = (
    record_count_ok
    and hash_integrity_ok
    and collision_consistency_ok
    and grouping_identity_ok
    and unique_hash_count_ok
    and duplicate_group_count_ok
    and duplicate_member_count_ok
    and source_member_counts_ok
    and relation_counts_ok
)

print(
    "CODE CELL 19 GATE:",
    "PASS"
    if cell_gate
    else "FAIL",
)
```

<h3>Discovery شباهت‌های Cross-source میان Website، Book و Transcript</h3>

<p>
تحلیل Near-duplicate تا این مرحله به‌صورت Source-aware و درون هر Source Type انجام شده است.
این رویکرد برای جلوگیری از مقایسه کور واحدهای ناهمگون ضروری بود، اما برای تکمیل Deduplication باید شباهت میان Source Typeهای متفاوت نیز بررسی شود.
</p>

<p>
در این مرحله سه جفت Source بررسی می‌شوند:
</p>

<ul>
<li>Website ↔ Book</li>
<li>Website ↔ Transcript</li>
<li>Book ↔ Transcript</li>
</ul>

<p>
هدف Cross-source analysis حذف یا ادغام منابع نیست.
شباهت میان دو Source مختلف می‌تواند ناشی از
بازنشر،
اقتباس،
نسخه دیگر،
نقل‌قول،
اشتراک مفهومی،
یا Evidence مستقل باشد.
بنابراین Source identity، Attribution و Lineage باید حفظ شوند.
</p>

<h4>روش Candidate Generation</h4>

<p>
برای هر جفت Source، Character TF-IDF با تنظیمات Benchmark قبلی روی مجموعه مشترک همان دو Source ساخته می‌شود.
سپس Documentهای هر طرف فقط در Source Type مقابل جست‌وجو می‌شوند.
</p>

<ul>
<li>Exact-text matchهای Cross-source کنار گذاشته می‌شوند، زیرا قبلاً جداگانه اعتبارسنجی شده‌اند.</li>
<li>برای هر Document نزدیک‌ترین Candidate غیرExact در Source مقابل ثبت می‌شود.</li>
<li>جست‌وجو در هر دو جهت انجام می‌شود و Pairهای تکراری در پایان یکی می‌شوند.</li>
<li>هیچ Similarity threshold تولیدی در این مرحله انتخاب نمی‌شود.</li>
</ul>

<h4>خروجی مورد انتظار</h4>

<ul>
<li>توزیع Similarity نزدیک‌ترین Cross-source Candidateها.</li>
<li>بالاترین Pairهای Website–Book، Website–Transcript و Book–Transcript.</li>
<li>Character TF-IDF، SequenceMatcher، Token Jaccard و Length Ratio برای Candidateهای برتر.</li>
<li>تفکیک Pairهای کوتاه از Pairهای دارای متن معنادارتر.</li>
</ul>

<h4>اصل مهم</h4>

<p>
حتی Similarity بسیار بالا بین دو Source مختلف به معنی Duplicate قابل حذف نیست.
این مرحله فقط Evidence ایجاد می‌کند تا بعداً Relationهایی مانند Reuse، Adaptation، Version یا Independent Evidence با احتیاط بررسی شوند.
</p>

<h4>معیار PASS</h4>

<p>
هر سه Source-pair باید بدون خطا تحلیل شوند،
تمام Candidateها باید واقعاً Cross-source باشند،
Self-pair و Exact-text pair نباید در خروجی باقی بماند،
و تمام Scoreهای محاسبه‌شده باید finite باشند.
</p>

</div>

```python
# Code Cell 20

from collections import Counter
from difflib import SequenceMatcher
import json
import math
import re

import numpy as np
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.neighbors import NearestNeighbors

CROSS_SOURCE_NEIGHBORS = 8

# -------------------------------------------------------------------
# Independent corpus reload
# -------------------------------------------------------------------

cross_source_records = []

with hash_input_path.open(
    "r",
    encoding="utf-8",
) as handle:
    for line in handle:
        record = json.loads(line)

        cross_source_records.append(
            {
                "document_id": record["document_id"],
                "source_type": record["source_type"],
                "source_id": record["source_id"],
                "title": record.get("title"),
                "text": record["text"].strip(),
                "text_sha256": record["text_sha256"],
            }
        )

records_by_source_type = {
    source_type: [
        record
        for record in cross_source_records
        if record["source_type"] == source_type
    ]
    for source_type in [
        "website",
        "book",
        "transcript",
    ]
}

source_pairs = [
    ("website", "book"),
    ("website", "transcript"),
    ("book", "transcript"),
]

# -------------------------------------------------------------------
# Helpers
# -------------------------------------------------------------------

def cross_token_set(text):
    return set(
        re.findall(
            r"[\w\u0600-\u06FF\u200c]+",
            text,
            flags=re.UNICODE,
        )
    )

def cross_token_jaccard(text_a, text_b):
    tokens_a = cross_token_set(text_a)
    tokens_b = cross_token_set(text_b)

    union = tokens_a | tokens_b

    if not union:
        return (
            1.0
            if text_a == text_b
            else 0.0
        )

    return (
        len(tokens_a & tokens_b)
        / len(union)
    )

def cross_length_ratio(text_a, text_b):
    length_a = len(text_a)
    length_b = len(text_b)

    maximum = max(
        length_a,
        length_b,
    )

    if maximum == 0:
        return 1.0

    return (
        min(
            length_a,
            length_b,
        )
        / maximum
    )

# -------------------------------------------------------------------
# Directional cross-source nearest-neighbor search
# -------------------------------------------------------------------

directional_cross_results = []

for source_a, source_b in source_pairs:
    records_a = records_by_source_type[
        source_a
    ]

    records_b = records_by_source_type[
        source_b
    ]

    combined_texts = [
        record["text"]
        for record in records_a
    ] + [
        record["text"]
        for record in records_b
    ]

    print("\n" + "=" * 100)
    print(
        "SOURCE PAIR:",
        source_a,
        "<->",
        source_b,
    )
    print(
        source_a,
        "documents:",
        len(records_a),
    )
    print(
        source_b,
        "documents:",
        len(records_b),
    )

    vectorizer = TfidfVectorizer(
        analyzer="char_wb",
        ngram_range=(3, 5),
        lowercase=False,
        min_df=1,
    )

    vectorizer.fit(
        combined_texts
    )

    print(
        "Vocabulary size:",
        len(
            vectorizer.vocabulary_
        ),
    )

    matrices = {
        source_a: vectorizer.transform(
            [
                record["text"]
                for record in records_a
            ]
        ),
        source_b: vectorizer.transform(
            [
                record["text"]
                for record in records_b
            ]
        ),
    }

    records_map = {
        source_a: records_a,
        source_b: records_b,
    }

    # Search both directions.
    for query_source, target_source in [
        (source_a, source_b),
        (source_b, source_a),
    ]:
        query_records = records_map[
            query_source
        ]

        target_records = records_map[
            target_source
        ]

        query_matrix = matrices[
            query_source
        ]

        target_matrix = matrices[
            target_source
        ]

        neighbor_count = min(
            CROSS_SOURCE_NEIGHBORS,
            len(target_records),
        )

        nn_model = NearestNeighbors(
            n_neighbors=neighbor_count,
            metric="cosine",
            algorithm="brute",
            n_jobs=-1,
        )

        nn_model.fit(
            target_matrix
        )

        distances, indices = (
            nn_model.kneighbors(
                query_matrix,
                return_distance=True,
            )
        )

        found_count = 0

        for query_index, query_record in enumerate(
            query_records
        ):
            selected = None

            for distance, neighbor_index in zip(
                distances[query_index],
                indices[query_index],
            ):
                target_record = (
                    target_records[
                        int(neighbor_index)
                    ]
                )

                # Exact text is already handled by
                # independently validated exact dedup.
                if (
                    query_record["text_sha256"]
                    == target_record[
                        "text_sha256"
                    ]
                ):
                    continue

                similarity = (
                    1.0
                    - float(distance)
                )

                selected = {
                    "source_pair": (
                        f"{source_a}<->{source_b}"
                    ),
                    "query_source_type": (
                        query_source
                    ),
                    "target_source_type": (
                        target_source
                    ),
                    "document_id": (
                        query_record[
                            "document_id"
                        ]
                    ),
                    "source_id": (
                        query_record[
                            "source_id"
                        ]
                    ),
                    "title": (
                        query_record[
                            "title"
                        ]
                    ),
                    "text_sha256": (
                        query_record[
                            "text_sha256"
                        ]
                    ),
                    "text": (
                        query_record[
                            "text"
                        ]
                    ),
                    "candidate_document_id": (
                        target_record[
                            "document_id"
                        ]
                    ),
                    "candidate_source_id": (
                        target_record[
                            "source_id"
                        ]
                    ),
                    "candidate_title": (
                        target_record[
                            "title"
                        ]
                    ),
                    "candidate_text_sha256": (
                        target_record[
                            "text_sha256"
                        ]
                    ),
                    "candidate_text": (
                        target_record[
                            "text"
                        ]
                    ),
                    "similarity": similarity,
                }

                break

            if selected is not None:
                directional_cross_results.append(
                    selected
                )

                found_count += 1

        print(
            f"{query_source} -> {target_source}:",
            found_count,
            "non-exact nearest candidates",
        )

# -------------------------------------------------------------------
# Unique unordered cross-source pairs
# -------------------------------------------------------------------

unique_cross_pairs = {}

for result in directional_cross_results:
    pair_key = tuple(
        sorted(
            [
                result["document_id"],
                result[
                    "candidate_document_id"
                ],
            ]
        )
    )

    existing = unique_cross_pairs.get(
        pair_key
    )

    if (
        existing is None
        or result["similarity"]
        > existing["similarity"]
    ):
        unique_cross_pairs[
            pair_key
        ] = result

cross_pair_results = list(
    unique_cross_pairs.values()
)

# -------------------------------------------------------------------
# Add secondary metrics
# -------------------------------------------------------------------

for result in cross_pair_results:
    text_a = result["text"]
    text_b = result[
        "candidate_text"
    ]

    result[
        "sequence_matcher"
    ] = SequenceMatcher(
        None,
        text_a,
        text_b,
        autojunk=False,
    ).ratio()

    result[
        "token_jaccard"
    ] = cross_token_jaccard(
        text_a,
        text_b,
    )

    result[
        "length_ratio"
    ] = cross_length_ratio(
        text_a,
        text_b,
    )

    result[
        "chars_a"
    ] = len(text_a)

    result[
        "chars_b"
    ] = len(text_b)

    result[
        "words_a"
    ] = len(
        text_a.split()
    )

    result[
        "words_b"
    ] = len(
        text_b.split()
    )

cross_pair_results.sort(
    key=lambda row: (
        -row["similarity"],
        row["source_pair"],
        row["document_id"],
    )
)

# -------------------------------------------------------------------
# Distribution summary
# -------------------------------------------------------------------

print(
    "\n=== CROSS-SOURCE NEAREST NON-EXACT PROFILE ==="
)

print(
    "Directional candidate records:",
    len(
        directional_cross_results
    ),
)

print(
    "Unique unordered pairs:",
    len(
        cross_pair_results
    ),
)

for source_pair in [
    "website<->book",
    "website<->transcript",
    "book<->transcript",
]:
    rows = [
        row
        for row in cross_pair_results
        if row["source_pair"]
        == source_pair
    ]

    values = np.asarray(
        [
            row["similarity"]
            for row in rows
        ],
        dtype=float,
    )

    print("\nSOURCE PAIR:", source_pair)
    print(
        "Unique candidate pairs:",
        len(rows),
    )

    if len(values) == 0:
        continue

    print(
        " Min:",
        round(
            float(
                np.min(values)
            ),
            6,
        ),
    )

    print(
        " Median:",
        round(
            float(
                np.median(values)
            ),
            6,
        ),
    )

    print(
        " P90:",
        round(
            float(
                np.percentile(
                    values,
                    90,
                )
            ),
            6,
        ),
    )

    print(
        " P95:",
        round(
            float(
                np.percentile(
                    values,
                    95,
                )
            ),
            6,
        ),
    )

    print(
        " P99:",
        round(
            float(
                np.percentile(
                    values,
                    99,
                )
            ),
            6,
        ),
    )

    print(
        " Max:",
        round(
            float(
                np.max(values)
            ),
            6,
        ),
    )

# -------------------------------------------------------------------
# Descriptive score-band counts
# Same profiling bands as earlier cells.
# -------------------------------------------------------------------

def cross_similarity_band(score):
    if score >= 0.98:
        return "0.98_PLUS"

    if score >= 0.95:
        return "0.95_TO_0.98"

    if score >= 0.90:
        return "0.90_TO_0.95"

    if score >= 0.80:
        return "0.80_TO_0.90"

    return "BELOW_0.80"

cross_band_counts = Counter(
    (
        row["source_pair"],
        cross_similarity_band(
            row["similarity"]
        ),
    )
    for row in cross_pair_results
)

print(
    "\n=== CROSS-SOURCE PROFILING BANDS ==="
)

for source_pair in [
    "website<->book",
    "website<->transcript",
    "book<->transcript",
]:
    print(
        "\nSOURCE PAIR:",
        source_pair,
    )

    for band in [
        "0.98_PLUS",
        "0.95_TO_0.98",
        "0.90_TO_0.95",
        "0.80_TO_0.90",
        "BELOW_0.80",
    ]:
        print(
            f"- {band}:",
            cross_band_counts.get(
                (
                    source_pair,
                    band,
                ),
                0,
            ),
        )

# -------------------------------------------------------------------
# Inspect highest cross-source candidates
# -------------------------------------------------------------------

print(
    "\n=== HIGHEST CROSS-SOURCE NON-EXACT CANDIDATES ==="
)

for index, row in enumerate(
    cross_pair_results[:80],
    start=1,
):
    print(
        "\n"
        + "-" * 100
    )

    print(
        "PAIR",
        index,
    )

    print(
        "Source pair:",
        row["source_pair"],
    )

    print(
        "Character TF-IDF:",
        round(
            row["similarity"],
            6,
        ),
    )

    print(
        "SequenceMatcher:",
        round(
            row[
                "sequence_matcher"
            ],
            6,
        ),
    )

    print(
        "Token Jaccard:",
        round(
            row[
                "token_jaccard"
            ],
            6,
        ),
    )

    print(
        "Length ratio:",
        round(
            row["length_ratio"],
            6,
        ),
    )

    print(
        "A chars/words:",
        row["chars_a"],
        "/",
        row["words_a"],
    )

    print(
        "B chars/words:",
        row["chars_b"],
        "/",
        row["words_b"],
    )

    print(
        "A:",
        row["document_id"],
    )

    print(
        "A source_id:",
        row["source_id"],
    )

    print(
        "A title:",
        row["title"],
    )

    print(
        "B:",
        row[
            "candidate_document_id"
        ],
    )

    print(
        "B source_id:",
        row[
            "candidate_source_id"
        ],
    )

    print(
        "B title:",
        row[
            "candidate_title"
        ],
    )

    print(
        "A text:",
        row["text"]
        .replace(
            "\n",
            " ",
        )[:700],
    )

    print(
        "B text:",
        row[
            "candidate_text"
        ]
        .replace(
            "\n",
            " ",
        )[:700],
    )

# -------------------------------------------------------------------
# Validation
# -------------------------------------------------------------------

all_cross_source = all(
    row["query_source_type"]
    != row["target_source_type"]
    for row in directional_cross_results
)

no_exact_cross_pairs = all(
    row["text_sha256"]
    != row[
        "candidate_text_sha256"
    ]
    for row in cross_pair_results
)

no_self_cross_pairs = all(
    row["document_id"]
    != row[
        "candidate_document_id"
    ]
    for row in cross_pair_results
)

all_cross_scores_finite = all(
    math.isfinite(
        row[metric]
    )
    for row in cross_pair_results
    for metric in [
        "similarity",
        "sequence_matcher",
        "token_jaccard",
        "length_ratio",
    ]
)

all_source_pairs_present = all(
    any(
        row["source_pair"]
        == source_pair
        for row in cross_pair_results
    )
    for source_pair in [
        "website<->book",
        "website<->transcript",
        "book<->transcript",
    ]
)

print("\nValidation:")

print(
    "All candidates are cross-source:",
    "PASS"
    if all_cross_source
    else "FAIL",
)

print(
    "Exact-text pairs excluded:",
    "PASS"
    if no_exact_cross_pairs
    else "FAIL",
)

print(
    "Self-pairs excluded:",
    "PASS"
    if no_self_cross_pairs
    else "FAIL",
)

print(
    "All cross-source metrics finite:",
    "PASS"
    if all_cross_scores_finite
    else "FAIL",
)

print(
    "All three source-pairs represented:",
    "PASS"
    if all_source_pairs_present
    else "FAIL",
)

cell_gate = (
    all_cross_source
    and no_exact_cross_pairs
    and no_self_cross_pairs
    and all_cross_scores_finite
    and all_source_pairs_present
)

print(
    "CODE CELL 20 GATE:",
    "PASS"
    if cell_gate
    else "FAIL",
)
```

اما در Source دیگر ممکن است ساختار Metadata متفاوتی داشته باشد.
Policy فقط بعد از مشاهده شواهد واقعی تعیین خواهد شد.

و گزارش Nested metadata و Text profile باید بدون خطا برای هر سه Source تولید شود.

```python
# Code Cell 21

from collections import Counter, defaultdict
from pathlib import Path
import json
import math

schema_path = (
    repo_root
    / "schemas/phase2/document.schema.json"
)

with schema_path.open(
    "r",
    encoding="utf-8",
) as handle:
    document_schema = json.load(handle)

schema_required_fields = document_schema.get(
    "required",
    [],
)

# -------------------------------------------------------------------
# Independent dataset reload
# -------------------------------------------------------------------

quality_records = []

with hash_input_path.open(
    "r",
    encoding="utf-8",
) as handle:
    for line in handle:
        quality_records.append(
            json.loads(line)
        )

print("=== QUALITY DISCOVERY INPUT ===")
print(
    "Dataset:",
    hash_input_path.relative_to(
        repo_root
    ),
)
print(
    "Schema:",
    schema_path.relative_to(
        repo_root
    ),
)
print(
    "Documents:",
    len(quality_records),
)

print("\nSchema required fields:")
for field in schema_required_fields:
    print(
        "-",
        field,
    )

# -------------------------------------------------------------------
# Generic value-state helper
# -------------------------------------------------------------------

def value_state(
    record,
    field,
):
    if field not in record:
        return "MISSING"

    value = record[field]

    if value is None:
        return "NULL"

    if isinstance(
        value,
        str,
    ):
        if value.strip() == "":
            return "EMPTY_STRING"

        return "NONEMPTY"

    if isinstance(
        value,
        (list, dict),
    ):
        if len(value) == 0:
            return "EMPTY_CONTAINER"

        return "NONEMPTY"

    return "NONEMPTY"

# -------------------------------------------------------------------
# Required-field completeness
# -------------------------------------------------------------------

print(
    "\n=== REQUIRED FIELD COMPLETENESS ==="
)

required_field_audit = {}

for source_type in [
    "website",
    "book",
    "transcript",
]:
    source_records = [
        record
        for record in quality_records
        if record["source_type"]
        == source_type
    ]

    print(
        "\nSOURCE:",
        source_type,
    )
    print(
        "Documents:",
        len(source_records),
    )

    required_field_audit[
        source_type
    ] = {}

    for field in schema_required_fields:
        counts = Counter(
            value_state(
                record,
                field,
            )
            for record in source_records
        )

        required_field_audit[
            source_type
        ][field] = dict(counts)

        problem_count = sum(
            counts.get(
                state,
                0,
            )
            for state in [
                "MISSING",
                "NULL",
                "EMPTY_STRING",
                "EMPTY_CONTAINER",
            ]
        )

        print(
            f"- {field}:",
            "problems =",
            problem_count,
            "| states =",
            dict(
                sorted(
                    counts.items()
                )
            ),
        )

# -------------------------------------------------------------------
# Discover actual nested-object keys
# -------------------------------------------------------------------

nested_objects = [
    "attribution",
    "rights",
    "locator",
    "lineage",
    "source_metadata",
]

nested_key_inventory = {}

print(
    "\n=== NESTED OBJECT KEY DISCOVERY ==="
)

for object_name in nested_objects:
    keys_by_source = defaultdict(
        set
    )

    invalid_object_count = Counter()

    for record in quality_records:
        source_type = record[
            "source_type"
        ]

        value = record.get(
            object_name
        )

        if isinstance(
            value,
            dict,
        ):
            keys_by_source[
                source_type
            ].update(
                value.keys()
            )
        else:
            invalid_object_count[
                source_type
            ] += 1

    nested_key_inventory[
        object_name
    ] = {
        source_type: sorted(
            keys_by_source[
                source_type
            ]
        )
        for source_type in [
            "website",
            "book",
            "transcript",
        ]
    }

    print(
        "\nOBJECT:",
        object_name,
    )

    for source_type in [
        "website",
        "book",
        "transcript",
    ]:
        print(
            f"- {source_type} keys:",
            nested_key_inventory[
                object_name
            ][source_type],
        )

        print(
            "  non-dict values:",
            invalid_object_count[
                source_type
            ],
        )

# -------------------------------------------------------------------
# Nested key coverage
# -------------------------------------------------------------------

nested_coverage = {}

print(
    "\n=== NESTED KEY COVERAGE ==="
)

for object_name in nested_objects:
    nested_coverage[
        object_name
    ] = {}

    print(
        "\nOBJECT:",
        object_name,
    )

    for source_type in [
        "website",
        "book",
        "transcript",
    ]:
        source_records = [
            record
            for record in quality_records
            if record["source_type"]
            == source_type
        ]

        discovered_keys = (
            nested_key_inventory[
                object_name
            ][source_type]
        )

        nested_coverage[
            object_name
        ][source_type] = {}

        print(
            "\n SOURCE:",
            source_type,
        )

        for key in discovered_keys:
            states = Counter()

            for record in source_records:
                nested = record.get(
                    object_name
                )

                if not isinstance(
                    nested,
                    dict,
                ):
                    states[
                        "OBJECT_NOT_DICT"
                    ] += 1
                    continue

                if key not in nested:
                    states[
                        "MISSING"
                    ] += 1
                    continue

                value = nested[
                    key
                ]

                if value is None:
                    states[
                        "NULL"
                    ] += 1

                elif isinstance(
                    value,
                    str,
                ) and value.strip() == "":
                    states[
                        "EMPTY_STRING"
                    ] += 1

                elif isinstance(
                    value,
                    (list, dict),
                ) and len(value) == 0:
                    states[
                        "EMPTY_CONTAINER"
                    ] += 1

                else:
                    states[
                        "NONEMPTY"
                    ] += 1

            nested_coverage[
                object_name
            ][source_type][
                key
            ] = dict(states)

            nonempty_count = states.get(
                "NONEMPTY",
                0,
            )

            print(
                f"  - {key}:",
                nonempty_count,
                "/",
                len(source_records),
                "nonempty",
                "| states =",
                dict(
                    sorted(
                        states.items()
                    )
                ),
            )

# -------------------------------------------------------------------
# Selected top-level categorical distributions
# based only on fields actually present
# -------------------------------------------------------------------

print(
    "\n=== TOP-LEVEL CATEGORICAL DISTRIBUTIONS ==="
)

for field in [
    "language",
    "content_origin",
]:
    print(
        "\nFIELD:",
        field,
    )

    for source_type in [
        "website",
        "book",
        "transcript",
    ]:
        values = Counter()

        for record in quality_records:
            if (
                record["source_type"]
                != source_type
            ):
                continue

            value = record.get(
                field
            )

            if isinstance(
                value,
                (dict, list),
            ):
                value = json.dumps(
                    value,
                    ensure_ascii=False,
                    sort_keys=True,
                )

            values[
                str(value)
            ] += 1

        print(
            f"- {source_type}:",
            dict(
                values.most_common()
            ),
        )

# -------------------------------------------------------------------
# Rights distributions from discovered keys
# -------------------------------------------------------------------

print(
    "\n=== RIGHTS VALUE DISTRIBUTIONS ==="
)

for source_type in [
    "website",
    "book",
    "transcript",
]:
    source_records = [
        record
        for record in quality_records
        if record["source_type"]
        == source_type
    ]

    rights_keys = nested_key_inventory[
        "rights"
    ][source_type]

    print(
        "\nSOURCE:",
        source_type,
    )

    for key in rights_keys:
        values = Counter()

        for record in source_records:
            rights = record.get(
                "rights"
            )

            if not isinstance(
                rights,
                dict,
            ):
                values[
                    "OBJECT_NOT_DICT"
                ] += 1
                continue

            if key not in rights:
                values[
                    "MISSING"
                ] += 1
                continue

            value = rights[
                key
            ]

            if isinstance(
                value,
                (dict, list),
            ):
                value = json.dumps(
                    value,
                    ensure_ascii=False,
                    sort_keys=True,
                )

            values[
                str(value)
            ] += 1

        print(
            f"- {key}:",
            dict(
                values.most_common(
                    20
                )
            ),
        )

# -------------------------------------------------------------------
# Text quality / length profile
# -------------------------------------------------------------------

print(
    "\n=== TEXT QUALITY PROFILE ==="
)

text_quality_summary = {}

for source_type in [
    "website",
    "book",
    "transcript",
]:
    source_records = [
        record
        for record in quality_records
        if record["source_type"]
        == source_type
    ]

    texts = [
        record.get(
            "text"
        )
        for record in source_records
    ]

    non_string_count = sum(
        not isinstance(
            text,
            str,
        )
        for text in texts
    )

    empty_string_count = sum(
        isinstance(
            text,
            str,
        )
        and text == ""
        for text in texts
    )

    whitespace_only_count = sum(
        isinstance(
            text,
            str,
        )
        and text != ""
        and text.strip() == ""
        for text in texts
    )

    valid_texts = [
        text
        for text in texts
        if isinstance(
            text,
            str,
        )
    ]

    char_lengths = np.asarray(
        [
            len(text)
            for text in valid_texts
        ],
        dtype=float,
    )

    word_lengths = np.asarray(
        [
            len(
                text.split()
            )
            for text in valid_texts
        ],
        dtype=float,
    )

    text_quality_summary[
        source_type
    ] = {
        "documents": len(
            source_records
        ),
        "non_string_text": (
            non_string_count
        ),
        "empty_string": (
            empty_string_count
        ),
        "whitespace_only": (
            whitespace_only_count
        ),
    }

    print(
        "\nSOURCE:",
        source_type,
    )

    print(
        "Documents:",
        len(source_records),
    )

    print(
        "Non-string text:",
        non_string_count,
    )

    print(
        "Empty text:",
        empty_string_count,
    )

    print(
        "Whitespace-only text:",
        whitespace_only_count,
    )

    if len(char_lengths):
        print(
            "Character length min/P25/median/P75/P95/max:",
            round(
                float(
                    np.min(
                        char_lengths
                    )
                ),
                2,
            ),
            "/",
            round(
                float(
                    np.percentile(
                        char_lengths,
                        25,
                    )
                ),
                2,
            ),
            "/",
            round(
                float(
                    np.median(
                        char_lengths
                    )
                ),
                2,
            ),
            "/",
            round(
                float(
                    np.percentile(
                        char_lengths,
                        75,
                    )
                ),
                2,
            ),
            "/",
            round(
                float(
                    np.percentile(
                        char_lengths,
                        95,
                    )
                ),
                2,
            ),
            "/",
            round(
                float(
                    np.max(
                        char_lengths
                    )
                ),
                2,
            ),
        )

        print(
            "Word length min/P25/median/P75/P95/max:",
            round(
                float(
                    np.min(
                        word_lengths
                    )
                ),
                2,
            ),
            "/",
            round(
                float(
                    np.percentile(
                        word_lengths,
                        25,
                    )
                ),
                2,
            ),
            "/",
            round(
                float(
                    np.median(
                        word_lengths
                    )
                ),
                2,
            ),
            "/",
            round(
                float(
                    np.percentile(
                        word_lengths,
                        75,
                    )
                ),
                2,
            ),
            "/",
            round(
                float(
                    np.percentile(
                        word_lengths,
                        95,
                    )
                ),
                2,
            ),
            "/",
            round(
                float(
                    np.max(
                        word_lengths
                    )
                ),
                2,
            ),
        )

# -------------------------------------------------------------------
# Validation
# -------------------------------------------------------------------

document_count_ok = (
    len(quality_records)
    == 19456
)

schema_required_present = (
    isinstance(
        schema_required_fields,
        list,
    )
    and len(
        schema_required_fields
    ) > 0
)

all_sources_present = (
    {
        record["source_type"]
        for record in quality_records
    }
    >= {
        "website",
        "book",
        "transcript",
    }
)

all_required_fields_audited = all(
    field
    in required_field_audit[
        source_type
    ]
    for source_type in [
        "website",
        "book",
        "transcript",
    ]
    for field in schema_required_fields
)

all_nested_objects_audited = all(
    object_name
    in nested_coverage
    for object_name
    in nested_objects
)

print(
    "\nValidation:"
)

print(
    "Document count:",
    "PASS"
    if document_count_ok
    else "FAIL",
)

print(
    "Schema required fields discovered:",
    "PASS"
    if schema_required_present
    else "FAIL",
)

print(
    "All three source types present:",
    "PASS"
    if all_sources_present
    else "FAIL",
)

print(
    "All required fields audited:",
    "PASS"
    if all_required_fields_audited
    else "FAIL",
)

print(
    "All nested objects audited:",
    "PASS"
    if all_nested_objects_audited
    else "FAIL",
)

cell_gate = (
    document_count_ok
    and schema_required_present
    and all_sources_present
    and all_required_fields_audited
    and all_nested_objects_audited
)

print(
    "CODE CELL 21 GATE:",
    "PASS"
    if cell_gate
    else "FAIL",
)
```

<h3>Discovery قواعد Schema و تفکیک Schema Validity از Quality Gap</h3>

<p>
Quality Discovery اولیه نشان داد که تعدادی از فیلدهای Required مقدار Null دارند.
اما Required بودن یک فیلد در JSON Schema فقط الزام وجود Key را مشخص می‌کند و لزوماً Null بودن مقدار را ممنوع نمی‌کند.
</p>

<p>
بنابراین پیش از طراحی Quality Score یا Quarantine Policy باید برای هر فیلد Required مشخص شود:
</p>

<ul>
<li>چه Typeهایی در Schema مجاز هستند.</li>
<li>آیا <code>null</code> صراحتاً مجاز است یا خیر.</li>
<li>آیا Enum، Pattern، Minimum Length یا محدودیت دیگری وجود دارد.</li>
<li>Nested objectها چه Required fieldهایی دارند.</li>
<li>آیا Attribution، Rights، Locator و Lineage محدودیت‌های Source-independent دارند یا خیر.</li>
</ul>

<p>
این تفکیک ضروری است زیرا یک مقدار Null ممکن است از نظر Schema کاملاً معتبر باشد ولی از نظر Quality یک Metadata gap محسوب شود.
برعکس، یک فیلد Non-null نیز ممکن است از نظر Policy کیفیت کافی نداشته باشد.
</p>

<h4>هدف این مرحله</h4>

<p>
ایجاد یک Contract واقعی از Schema برای فیلدهای اصلی و Nested objectهای مهم است تا سه مفهوم از یکدیگر جدا بمانند:
</p>

<ul>
<li><code>SCHEMA_INVALID</code>: مقدار با Schema ناسازگار است.</li>
<li><code>SCHEMA_VALID_BUT_INCOMPLETE</code>: مقدار طبق Schema مجاز است ولی ممکن است برای Quality یا Release کافی نباشد.</li>
<li><code>QUALITY_ACCEPTABLE</code>: فقط بعد از بررسی Policy و Evidence قابل تصمیم‌گیری است.</li>
</ul>

<h4>اصل مهم</h4>

<p>
در این مرحله هیچ Quality Score، Release Gate یا Quarantine Rule ساخته نمی‌شود.
فقط Contract خود Schema استخراج می‌شود.
</p>

Schema باید برای تمام فیلدهای Required و Nested objectهای
<code>attribution</code>،

<code>lineage</code>

قابل تفسیر باشد و Nullability و Nested Required fields به‌صورت صریح گزارش شوند.

```python
# Code Cell 22

from collections import Counter
import json

# -------------------------------------------------------------------
# Schema helper functions
# -------------------------------------------------------------------

def resolve_local_ref(schema_root, node):
    if not isinstance(node, dict):
        return node

    ref = node.get("$ref")

    if not ref:
        return node

    if not ref.startswith("#/"):
        return node

    target = schema_root

    for part in ref[2:].split("/"):
        target = target[part]

    return target


def json_type_of_literal(value):
    if value is None:
        return "null"

    if isinstance(value, bool):
        return "boolean"

    if isinstance(value, str):
        return "string"

    if isinstance(value, int):
        return "integer"

    if isinstance(value, float):
        return "number"

    if isinstance(value, list):
        return "array"

    if isinstance(value, dict):
        return "object"

    return None


def collect_allowed_types(
    schema_root,
    node,
):
    """
    Infer allowed JSON types from:
    - type
    - enum
    - const
    - anyOf
    - oneOf
    - resolved local $ref

    This is necessary because JSON Schema may legally permit
    null through enum/const without declaring type: null.
    """

    node = resolve_local_ref(
        schema_root,
        node,
    )

    types = set()

    if not isinstance(node, dict):
        return types

    node_type = node.get("type")

    if isinstance(node_type, str):
        types.add(node_type)

    elif isinstance(node_type, list):
        types.update(node_type)

    if "enum" in node:
        for value in node["enum"]:
            inferred_type = (
                json_type_of_literal(
                    value
                )
            )

            if inferred_type is not None:
                types.add(
                    inferred_type
                )

    if "const" in node:
        inferred_type = (
            json_type_of_literal(
                node["const"]
            )
        )

        if inferred_type is not None:
            types.add(
                inferred_type
            )

    for keyword in [
        "anyOf",
        "oneOf",
    ]:
        for option in node.get(
            keyword,
            [],
        ):
            types.update(
                collect_allowed_types(
                    schema_root,
                    option,
                )
            )

    return types


def collect_constraint_summary(
    schema_root,
    node,
):
    node = resolve_local_ref(
        schema_root,
        node,
    )

    if not isinstance(node, dict):
        return {}

    summary = {}

    for key in [
        "enum",
        "const",
        "pattern",
        "format",
        "minLength",
        "maxLength",
        "minimum",
        "maximum",
        "minItems",
        "maxItems",
        "uniqueItems",
        "additionalProperties",
    ]:
        if key in node:
            summary[key] = node[key]

    if "anyOf" in node:
        summary["anyOf_count"] = len(
            node["anyOf"]
        )

    if "oneOf" in node:
        summary["oneOf_count"] = len(
            node["oneOf"]
        )

    return summary


def get_object_contract(
    schema_root,
    node,
):
    node = resolve_local_ref(
        schema_root,
        node,
    )

    if not isinstance(node, dict):
        return {
            "required": [],
            "properties": {},
        }

    candidates = [node]

    for keyword in [
        "anyOf",
        "oneOf",
    ]:
        candidates.extend(
            node.get(
                keyword,
                []
            )
        )

    required = set()
    properties = {}

    for candidate in candidates:
        candidate = resolve_local_ref(
            schema_root,
            candidate,
        )

        if not isinstance(
            candidate,
            dict,
        ):
            continue

        required.update(
            candidate.get(
                "required",
                []
            )
        )

        for key, value in candidate.get(
            "properties",
            {},
        ).items():
            properties[
                key
            ] = value

    return {
        "required": sorted(
            required
        ),
        "properties": properties,
    }


def schema_allows_null(
    schema_root,
    node,
):
    return (
        "null"
        in collect_allowed_types(
            schema_root,
            node,
        )
    )


# -------------------------------------------------------------------
# Top-level property discovery
# -------------------------------------------------------------------

top_level_properties = (
    document_schema.get(
        "properties",
        {}
    )
)

print(
    "=== TOP-LEVEL REQUIRED FIELD SCHEMA CONTRACT ==="
)

required_schema_contract = {}

for field in schema_required_fields:
    node = top_level_properties.get(
        field,
        {}
    )

    allowed_types = sorted(
        collect_allowed_types(
            document_schema,
            node,
        )
    )

    constraints = (
        collect_constraint_summary(
            document_schema,
            node,
        )
    )

    nullable = schema_allows_null(
        document_schema,
        node,
    )

    required_schema_contract[
        field
    ] = {
        "allowed_types": allowed_types,
        "nullable": nullable,
        "constraints": constraints,
    }

    print(
        "\nFIELD:",
        field,
    )

    print(
        " Allowed types:",
        allowed_types,
    )

    print(
        " Nullable:",
        nullable,
    )

    print(
        " Constraints:",
        constraints,
    )

# -------------------------------------------------------------------
# Nested-object schema contracts
# -------------------------------------------------------------------

schema_nested_objects = [
    "attribution",
    "rights",
    "locator",
    "lineage",
    "source_metadata",
]

nested_schema_contracts = {}

print(
    "\n=== NESTED OBJECT SCHEMA CONTRACTS ==="
)

for object_name in schema_nested_objects:
    node = top_level_properties.get(
        object_name,
        {}
    )

    object_contract = (
        get_object_contract(
            document_schema,
            node,
        )
    )

    nested_schema_contracts[
        object_name
    ] = object_contract

    print(
        "\nOBJECT:",
        object_name,
    )

    print(
        " Required nested keys:",
        object_contract[
            "required"
        ],
    )

    print(
        " Property count:",
        len(
            object_contract[
                "properties"
            ]
        ),
    )

    for key, property_node in sorted(
        object_contract[
            "properties"
        ].items()
    ):
        allowed_types = sorted(
            collect_allowed_types(
                document_schema,
                property_node,
            )
        )

        constraints = (
            collect_constraint_summary(
                document_schema,
                property_node,
            )
        )

        nullable = schema_allows_null(
            document_schema,
            property_node,
        )

        print(
            f" - {key}:",
        )

        print(
            "   types:",
            allowed_types,
        )

        print(
            "   nullable:",
            nullable,
        )

        print(
            "   constraints:",
            constraints,
        )

# -------------------------------------------------------------------
# Compare observed NULLs with schema nullability
# -------------------------------------------------------------------

print(
    "\n=== OBSERVED NULLS VS SCHEMA NULLABILITY ==="
)

schema_nullability_findings = []

for source_type in [
    "website",
    "book",
    "transcript",
]:
    source_records = [
        record
        for record in quality_records
        if record["source_type"]
        == source_type
    ]

    print(
        "\nSOURCE:",
        source_type,
    )

    for field in schema_required_fields:
        null_count = sum(
            record.get(
                field
            ) is None
            for record in source_records
        )

        if null_count == 0:
            continue

        nullable = (
            required_schema_contract[
                field
            ][
                "nullable"
            ]
        )

        finding = {
            "source_type": (
                source_type
            ),
            "field": field,
            "null_count": (
                null_count
            ),
            "schema_nullable": (
                nullable
            ),
        }

        schema_nullability_findings.append(
            finding
        )

        print(
            "-",
            field,
            "| null count:",
            null_count,
            "| schema nullable:",
            nullable,
        )

# -------------------------------------------------------------------
# Check nested required-key presence against observed data
# -------------------------------------------------------------------

print(
    "\n=== NESTED REQUIRED-KEY PRESENCE ==="
)

nested_required_audit = {}

for object_name in schema_nested_objects:
    required_keys = (
        nested_schema_contracts[
            object_name
        ][
            "required"
        ]
    )

    nested_required_audit[
        object_name
    ] = {}

    print(
        "\nOBJECT:",
        object_name,
    )

    if not required_keys:
        print(
            " No nested required keys declared."
        )
        continue

    for source_type in [
        "website",
        "book",
        "transcript",
    ]:
        source_records = [
            record
            for record in quality_records
            if record[
                "source_type"
            ]
            == source_type
        ]

        key_counts = {}

        for key in required_keys:
            missing_count = 0

            for record in source_records:
                nested = record.get(
                    object_name
                )

                if (
                    not isinstance(
                        nested,
                        dict,
                    )
                    or key not in nested
                ):
                    missing_count += 1

            key_counts[
                key
            ] = missing_count

        nested_required_audit[
            object_name
        ][
            source_type
        ] = key_counts

        print(
            " SOURCE:",
            source_type,
        )

        for key in required_keys:
            print(
                f"  - {key}:",
                "missing =",
                key_counts[
                    key
                ],
            )

# -------------------------------------------------------------------
# Nested observed NULL validity
# -------------------------------------------------------------------

print(
    "\n=== NESTED OBSERVED NULLS VS SCHEMA NULLABILITY ==="
)

nested_invalid_null_findings = []

for object_name in [
    "attribution",
    "rights",
    "lineage",
]:
    contract = (
        nested_schema_contracts[
            object_name
        ]
    )

    for key, property_node in contract[
        "properties"
    ].items():
        nullable = schema_allows_null(
            document_schema,
            property_node,
        )

        for source_type in [
            "website",
            "book",
            "transcript",
        ]:
            null_count = 0

            for record in quality_records:
                if (
                    record["source_type"]
                    != source_type
                ):
                    continue

                nested = record.get(
                    object_name
                )

                if (
                    isinstance(
                        nested,
                        dict,
                    )
                    and key in nested
                    and nested[key] is None
                ):
                    null_count += 1

            if null_count == 0:
                continue

            print(
                "-",
                object_name,
                ".",
                key,
                "| source:",
                source_type,
                "| null count:",
                null_count,
                "| schema nullable:",
                nullable,
            )

            if not nullable:
                nested_invalid_null_findings.append(
                    {
                        "object": (
                            object_name
                        ),
                        "field": key,
                        "source_type": (
                            source_type
                        ),
                        "null_count": (
                            null_count
                        ),
                    }
                )

# -------------------------------------------------------------------
# Schema-level interpretation summary
# -------------------------------------------------------------------

print(
    "\n=== SCHEMA NULLABILITY SUMMARY ==="
)

nullable_required_fields = [
    field
    for field, contract
    in required_schema_contract.items()
    if contract["nullable"]
]

nonnullable_required_fields = [
    field
    for field, contract
    in required_schema_contract.items()
    if not contract["nullable"]
]

print(
    "Required fields allowing null:",
    nullable_required_fields,
)

print(
    "Required fields not allowing null:",
    nonnullable_required_fields,
)

schema_invalid_observed_nulls = [
    finding
    for finding
    in schema_nullability_findings
    if not finding[
        "schema_nullable"
    ]
]

print(
    "Observed top-level NULL populations violating schema:",
    len(
        schema_invalid_observed_nulls
    ),
)

for finding in schema_invalid_observed_nulls:
    print(
        finding
    )

print(
    "Observed nested NULL populations violating schema:",
    len(
        nested_invalid_null_findings
    ),
)

for finding in nested_invalid_null_findings:
    print(
        finding
    )

# -------------------------------------------------------------------
# Specific semantic sanity checks for enum/const inference
# -------------------------------------------------------------------

content_origin_contract = (
    required_schema_contract[
        "content_origin"
    ]
)

content_origin_enum = (
    content_origin_contract[
        "constraints"
    ].get(
        "enum",
        [],
    )
)

content_origin_null_consistency = (
    (
        None
        in content_origin_enum
    )
    == content_origin_contract[
        "nullable"
    ]
)

source_type_contract = (
    required_schema_contract[
        "source_type"
    ]
)

schema_version_contract = (
    required_schema_contract[
        "schema_version"
    ]
)

enum_const_type_inference_ok = (
    "string"
    in source_type_contract[
        "allowed_types"
    ]
    and "string"
    in schema_version_contract[
        "allowed_types"
    ]
)

# -------------------------------------------------------------------
# Validation
# -------------------------------------------------------------------

all_required_fields_resolved = (
    set(
        required_schema_contract
    )
    == set(
        schema_required_fields
    )
)

all_nested_objects_resolved = (
    set(
        nested_schema_contracts
    )
    == set(
        schema_nested_objects
    )
)

nullability_defined_for_all = all(
    isinstance(
        contract["nullable"],
        bool,
    )
    for contract
    in required_schema_contract.values()
)

nested_required_audit_complete = all(
    object_name
    in nested_required_audit
    for object_name
    in schema_nested_objects
)

top_level_nulls_schema_valid = (
    len(
        schema_invalid_observed_nulls
    )
    == 0
)

nested_nulls_schema_valid = (
    len(
        nested_invalid_null_findings
    )
    == 0
)

print(
    "\nValidation:"
)

print(
    "All required fields resolved:",
    "PASS"
    if all_required_fields_resolved
    else "FAIL",
)

print(
    "All nested object contracts resolved:",
    "PASS"
    if all_nested_objects_resolved
    else "FAIL",
)

print(
    "Nullability determined for every required field:",
    "PASS"
    if nullability_defined_for_all
    else "FAIL",
)

print(
    "Nested required-key audit complete:",
    "PASS"
    if nested_required_audit_complete
    else "FAIL",
)

print(
    "Enum/const type inference:",
    "PASS"
    if enum_const_type_inference_ok
    else "FAIL",
)

print(
    "content_origin enum/null consistency:",
    "PASS"
    if content_origin_null_consistency
    else "FAIL",
)

print(
    "Observed top-level NULLs allowed by schema:",
    "PASS"
    if top_level_nulls_schema_valid
    else "FAIL",
)

print(
    "Observed nested NULLs allowed by schema:",
    "PASS"
    if nested_nulls_schema_valid
    else "FAIL",
)

cell_gate = (
    all_required_fields_resolved
    and all_nested_objects_resolved
    and nullability_defined_for_all
    and nested_required_audit_complete
    and enum_const_type_inference_ok
    and content_origin_null_consistency
    and top_level_nulls_schema_valid
    and nested_nulls_schema_valid
)

print(
    "CODE CELL 22 GATE:",
    "PASS"
    if cell_gate
    else "FAIL",
)
```

<h3>Discovery معنایی Attribution، Rights و Locator برای Quality Policy</h3>

<p>
Schema validation فقط مشخص می‌کند که یک Document از نظر ساختاری مجاز است.
اما برای Quality Policy باید بررسی شود که Metadata موجود واقعاً چه میزان Evidence قابل استفاده ایجاد می‌کند.
</p>

<p>
برای مثال در Transcript تعداد
<code>5,016</code>
Document دارای <code>attribution.speaker = null</code> هستند،
در حالی که <code>source_metadata.speaker</code> برای تمام Transcriptها مقدار دارد.
قبل از اینکه Missing speaker به‌عنوان Quality failure یا Quarantine condition تعریف شود،
باید رابطه واقعی این دو فیلد بررسی شود.
</p>

<p>
همین مسئله درباره Rights نیز وجود دارد.
وجود <code>EVIDENCE_REQUIRED</code> یک وضعیت صریح است،
اما Null بودن Rights در Website یا نبود <code>rights_evidence_id</code> هنوز بدون بررسی Evidence و الگوی Source نباید به‌عنوان Failure نهایی تفسیر شود.
</p>

<h4>بررسی‌های این مرحله</h4>

<ul>
<li>Type و مقدار واقعی Attribution fieldهای کلیدی به تفکیک Source.</li>
<li>رابطه <code>attribution.speaker</code> با <code>source_metadata.speaker</code> در Transcript.</li>
<li>نمونه‌های واقعی Transcriptهایی که Attribution speaker ندارند.</li>
<li>توزیع <code>attribution_status</code> و <code>content_authority_type</code>.</li>
<li>ترکیب واقعی وضعیت‌های Rights به تفکیک Source.</li>
<li>Coverage و یکتایی Locatorهای اصلی هر Source.</li>
<li>بررسی اینکه Locatorهای موجود واقعاً Documentها را از یکدیگر متمایز می‌کنند.</li>
<li>بررسی سازگاری <code>lineage.raw_source_hash</code> با <code>raw_content_hash</code>.</li>
</ul>

<h4>اصل مهم</h4>

<p>
در این مرحله هنوز هیچ Quality Score یا Quarantine Rule ساخته نمی‌شود.
هدف این است که مشخص شود کدام Metadata gap واقعاً فقدان Evidence است،
کدام مقدار فقط Nullable و طبیعی است،
و کدام وضعیت ممکن است در مرحله Release به Gate جداگانه نیاز داشته باشد.
</p>

<h4>معیار PASS</h4>

<p>
تمام بررسی‌ها باید روی Dataset کامل انجام شوند،
رابطه Speaker در Transcript باید بدون حدس گزارش شود،
Locator uniqueness باید برای هر Source اندازه‌گیری شود
و ناسازگاری Hash در Lineage، در صورت وجود، باید صریحاً گزارش شود.
</p>

```python
# Code Cell 23

from collections import Counter, defaultdict
import json
import math

# -------------------------------------------------------------------
# Helpers
# -------------------------------------------------------------------

def stable_value_repr(value):
    if isinstance(
        value,
        (dict, list),
    ):
        return json.dumps(
            value,
            ensure_ascii=False,
            sort_keys=True,
        )

    return repr(value)


def value_type_name(value):
    if value is None:
        return "null"

    return type(value).__name__


# -------------------------------------------------------------------
# Attribution semantic profile
# -------------------------------------------------------------------

print("=== ATTRIBUTION SEMANTIC PROFILE ===")

for source_type in [
    "website",
    "book",
    "transcript",
]:
    source_records = [
        record
        for record in quality_records
        if record["source_type"]
        == source_type
    ]

    print("\nSOURCE:", source_type)
    print(
        "Documents:",
        len(source_records),
    )

    for field in [
        "attribution_status",
        "content_authority_type",
        "author",
        "knowledge_owner",
        "speaker",
    ]:
        type_counts = Counter()
        value_counts = Counter()

        for record in source_records:
            attribution = record[
                "attribution"
            ]

            value = attribution.get(
                field
            )

            type_counts[
                value_type_name(
                    value
                )
            ] += 1

            value_counts[
                stable_value_repr(
                    value
                )
            ] += 1

        print(
            f"\n- {field}"
        )
        print(
            "  Types:",
            dict(
                sorted(
                    type_counts.items()
                )
            ),
        )
        print(
            "  Top values:",
            value_counts.most_common(
                15
            ),
        )

# -------------------------------------------------------------------
# Transcript speaker relationship
# -------------------------------------------------------------------

print(
    "\n=== TRANSCRIPT SPEAKER RELATIONSHIP ==="
)

transcript_records = [
    record
    for record in quality_records
    if record["source_type"]
    == "transcript"
]

speaker_relation_counts = Counter()

speaker_relation_examples = defaultdict(
    list
)

for record in transcript_records:
    attribution_speaker = (
        record["attribution"].get(
            "speaker"
        )
    )

    source_metadata_speaker = (
        record["source_metadata"].get(
            "speaker"
        )
    )

    if (
        attribution_speaker is None
        and source_metadata_speaker
        is None
    ):
        relation = (
            "BOTH_NULL"
        )

    elif (
        attribution_speaker is None
        and source_metadata_speaker
        is not None
    ):
        relation = (
            "ATTRIBUTION_NULL_METADATA_PRESENT"
        )

    elif (
        attribution_speaker
        is not None
        and source_metadata_speaker
        is None
    ):
        relation = (
            "ATTRIBUTION_PRESENT_METADATA_NULL"
        )

    elif (
        attribution_speaker
        == source_metadata_speaker
    ):
        relation = (
            "EXACT_VALUE_MATCH"
        )

    else:
        relation = (
            "BOTH_PRESENT_DIFFERENT"
        )

    speaker_relation_counts[
        relation
    ] += 1

    if len(
        speaker_relation_examples[
            relation
        ]
    ) < 15:
        speaker_relation_examples[
            relation
        ].append(
            {
                "document_id": record[
                    "document_id"
                ],
                "source_id": record[
                    "source_id"
                ],
                "attribution_speaker": (
                    attribution_speaker
                ),
                "source_metadata_speaker": (
                    source_metadata_speaker
                ),
                "source_metadata_record_type": (
                    record[
                        "source_metadata"
                    ].get(
                        "record_type"
                    )
                ),
                "text": record[
                    "text"
                ][:300],
            }
        )

print(
    "Speaker relation counts:",
    dict(
        sorted(
            speaker_relation_counts.items()
        )
    ),
)

print(
    "\nSpeaker relation examples:"
)

for relation in sorted(
    speaker_relation_examples
):
    print(
        "\nRELATION:",
        relation,
    )

    for example in (
        speaker_relation_examples[
            relation
        ]
    ):
        print(
            "-",
            example,
        )

# -------------------------------------------------------------------
# Transcript source_metadata.speaker actual value profile
# -------------------------------------------------------------------

print(
    "\n=== TRANSCRIPT SOURCE_METADATA.SPEAKER PROFILE ==="
)

metadata_speaker_types = Counter()
metadata_speaker_values = Counter()

for record in transcript_records:
    value = record[
        "source_metadata"
    ].get(
        "speaker"
    )

    metadata_speaker_types[
        value_type_name(
            value
        )
    ] += 1

    metadata_speaker_values[
        stable_value_repr(
            value
        )
    ] += 1

print(
    "Types:",
    dict(
        sorted(
            metadata_speaker_types.items()
        )
    ),
)

print(
    "Top values:",
    metadata_speaker_values.most_common(
        40
    ),
)

# -------------------------------------------------------------------
# Rights semantic combinations
# -------------------------------------------------------------------

print(
    "\n=== RIGHTS COMBINATION PROFILE ==="
)

rights_combination_counts = {}

for source_type in [
    "website",
    "book",
    "transcript",
]:
    source_records = [
        record
        for record in quality_records
        if record["source_type"]
        == source_type
    ]

    combinations = Counter()

    for record in source_records:
        rights = record[
            "rights"
        ]

        key = (
            stable_value_repr(
                rights.get(
                    "status"
                )
            ),
            stable_value_repr(
                rights.get(
                    "basis"
                )
            ),
            stable_value_repr(
                rights.get(
                    "rights_evidence_id"
                )
            ),
            stable_value_repr(
                rights.get(
                    "rights_holder"
                )
            ),
            stable_value_repr(
                rights.get(
                    "allowed_uses"
                )
            ),
        )

        combinations[
            key
        ] += 1

    rights_combination_counts[
        source_type
    ] = combinations

    print(
        "\nSOURCE:",
        source_type,
    )

    print(
        "Distinct rights combinations:",
        len(combinations),
    )

    for combination, count in (
        combinations.most_common(
            20
        )
    ):
        print(
            "- count:",
            count,
        )
        print(
            "  status:",
            combination[0],
        )
        print(
            "  basis:",
            combination[1],
        )
        print(
            "  evidence_id:",
            combination[2],
        )
        print(
            "  holder:",
            combination[3],
        )
        print(
            "  allowed_uses:",
            combination[4],
        )

# -------------------------------------------------------------------
# Locator uniqueness and coverage
# -------------------------------------------------------------------

print(
    "\n=== LOCATOR UNIQUENESS PROFILE ==="
)

locator_profile = {}

for source_type in [
    "website",
    "book",
    "transcript",
]:
    source_records = [
        record
        for record in quality_records
        if record["source_type"]
        == source_type
    ]

    serialized_locators = []

    for record in source_records:
        locator = record[
            "locator"
        ]

        serialized = json.dumps(
            locator,
            ensure_ascii=False,
            sort_keys=True,
            separators=(
                ",",
                ":",
            ),
        )

        serialized_locators.append(
            serialized
        )

    locator_counts = Counter(
        serialized_locators
    )

    duplicate_locator_groups = {
        locator: count
        for locator, count
        in locator_counts.items()
        if count > 1
    }

    duplicate_locator_members = sum(
        duplicate_locator_groups.values()
    )

    locator_profile[
        source_type
    ] = {
        "documents": len(
            source_records
        ),
        "unique_locators": len(
            locator_counts
        ),
        "duplicate_locator_groups": len(
            duplicate_locator_groups
        ),
        "duplicate_locator_members": (
            duplicate_locator_members
        ),
    }

    print(
        "\nSOURCE:",
        source_type,
    )

    print(
        "Documents:",
        len(source_records),
    )

    print(
        "Unique full locator objects:",
        len(locator_counts),
    )

    print(
        "Duplicate full-locator groups:",
        len(
            duplicate_locator_groups
        ),
    )

    print(
        "Duplicate full-locator members:",
        duplicate_locator_members,
    )

    if duplicate_locator_groups:
        print(
            "First duplicate locator groups:"
        )

        shown = 0

        for locator, count in sorted(
            duplicate_locator_groups.items(),
            key=lambda item: (
                -item[1],
                item[0],
            ),
        ):
            print(
                "- count:",
                count,
                "| locator:",
                locator[:700],
            )

            shown += 1

            if shown >= 15:
                break

# -------------------------------------------------------------------
# Source-specific locator identity fields
# -------------------------------------------------------------------

print(
    "\n=== SOURCE-SPECIFIC LOCATOR IDENTITY CHECKS ==="
)

source_specific_locator_checks = {}

# Website canonical_url
website_records = [
    record
    for record in quality_records
    if record["source_type"]
    == "website"
]

website_canonical_urls = [
    record["locator"].get(
        "canonical_url"
    )
    for record in website_records
]

website_canonical_url_counts = Counter(
    website_canonical_urls
)

source_specific_locator_checks[
    "website"
] = {
    "documents": len(
        website_records
    ),
    "nonempty_canonical_url": sum(
        isinstance(
            value,
            str,
        )
        and value.strip() != ""
        for value in website_canonical_urls
    ),
    "unique_canonical_url": len(
        {
            value
            for value
            in website_canonical_urls
            if value is not None
        }
    ),
    "duplicate_canonical_url_groups": sum(
        count > 1
        for count
        in website_canonical_url_counts.values()
    ),
}

print(
    "Website:",
    source_specific_locator_checks[
        "website"
    ],
)

# Book paragraph identity
book_records = [
    record
    for record in quality_records
    if record["source_type"]
    == "book"
]

book_primary_locator_keys = []

for record in book_records:
    locator = record[
        "locator"
    ]

    primary_key = (
        locator.get(
            "edition_id"
        ),
        locator.get(
            "paragraph_id"
        ),
        locator.get(
            "content_id"
        ),
        locator.get(
            "paragraph_number"
        ),
    )

    book_primary_locator_keys.append(
        primary_key
    )

book_primary_counts = Counter(
    book_primary_locator_keys
)

source_specific_locator_checks[
    "book"
] = {
    "documents": len(
        book_records
    ),
    "unique_primary_locator_keys": len(
        book_primary_counts
    ),
    "duplicate_primary_locator_groups": sum(
        count > 1
        for count
        in book_primary_counts.values()
    ),
}

print(
    "Book:",
    source_specific_locator_checks[
        "book"
    ],
)

# Transcript locator_id
transcript_locator_ids = [
    record["locator"].get(
        "locator_id"
    )
    for record in transcript_records
]

transcript_locator_id_counts = Counter(
    transcript_locator_ids
)

source_specific_locator_checks[
    "transcript"
] = {
    "documents": len(
        transcript_records
    ),
    "nonempty_locator_id": sum(
        isinstance(
            value,
            str,
        )
        and value.strip() != ""
        for value
        in transcript_locator_ids
    ),
    "unique_locator_id": len(
        {
            value
            for value
            in transcript_locator_ids
            if value is not None
        }
    ),
    "duplicate_locator_id_groups": sum(
        count > 1
        for count
        in transcript_locator_id_counts.values()
    ),
}

print(
    "Transcript:",
    source_specific_locator_checks[
        "transcript"
    ],
)

# -------------------------------------------------------------------
# Lineage raw hash vs raw_content_hash
# -------------------------------------------------------------------

print(
    "\n=== LINEAGE HASH CONSISTENCY ==="
)

lineage_hash_match_counts = Counter()
lineage_hash_mismatch_examples = []

for record in quality_records:
    lineage_hash = (
        record["lineage"].get(
            "raw_source_hash"
        )
    )

    raw_content_hash = (
        record.get(
            "raw_content_hash"
        )
    )

    if lineage_hash == raw_content_hash:
        relation = "MATCH"
    else:
        relation = "DIFFERENT"

    lineage_hash_match_counts[
        (
            record["source_type"],
            relation,
        )
    ] += 1

    if (
        relation == "DIFFERENT"
        and len(
            lineage_hash_mismatch_examples
        ) < 30
    ):
        lineage_hash_mismatch_examples.append(
            {
                "document_id": record[
                    "document_id"
                ],
                "source_type": record[
                    "source_type"
                ],
                "source_id": record[
                    "source_id"
                ],
                "raw_content_hash": (
                    raw_content_hash
                ),
                "lineage_raw_source_hash": (
                    lineage_hash
                ),
            }
        )

for source_type in [
    "website",
    "book",
    "transcript",
]:
    print(
        source_type,
        "| MATCH:",
        lineage_hash_match_counts.get(
            (
                source_type,
                "MATCH",
            ),
            0,
        ),
        "| DIFFERENT:",
        lineage_hash_match_counts.get(
            (
                source_type,
                "DIFFERENT",
            ),
            0,
        ),
    )

if lineage_hash_mismatch_examples:
    print(
        "\nFirst lineage hash differences:"
    )

    for example in (
        lineage_hash_mismatch_examples
    ):
        print(
            "-",
            example,
        )

# -------------------------------------------------------------------
# Validation
# -------------------------------------------------------------------

speaker_population_accounted = (
    sum(
        speaker_relation_counts.values()
    )
    == len(
        transcript_records
    )
)

rights_population_accounted = all(
    sum(
        combinations.values()
    )
    == sum(
        record["source_type"]
        == source_type
        for record in quality_records
    )
    for source_type, combinations
    in rights_combination_counts.items()
)

locator_population_accounted = all(
    locator_profile[
        source_type
    ][
        "documents"
    ]
    == sum(
        record["source_type"]
        == source_type
        for record in quality_records
    )
    for source_type in [
        "website",
        "book",
        "transcript",
    ]
)

lineage_population_accounted = (
    sum(
        lineage_hash_match_counts.values()
    )
    == len(
        quality_records
    )
)

print(
    "\nValidation:"
)

print(
    "Transcript speaker population accounted:",
    "PASS"
    if speaker_population_accounted
    else "FAIL",
)

print(
    "Rights population accounted:",
    "PASS"
    if rights_population_accounted
    else "FAIL",
)

print(
    "Locator population accounted:",
    "PASS"
    if locator_population_accounted
    else "FAIL",
)

print(
    "Lineage hash population accounted:",
    "PASS"
    if lineage_population_accounted
    else "FAIL",
)

cell_gate = (
    speaker_population_accounted
    and rights_population_accounted
    and locator_population_accounted
    and lineage_population_accounted
)

print(
    "CODE CELL 23 GATE:",
    "PASS"
    if cell_gate
    else "FAIL",
)
```

<h3>تحلیل معنایی Speaker Attribution در Transcript</h3>

<p>
Discovery قبلی نشان داد که
<code>source_metadata.speaker</code>
برای تمام Transcriptها یک Object ساختاری است و نباید مستقیماً با
<code>attribution.speaker</code>
که یک مقدار ساده Canonical است مقایسه شود.
</p>

<p>
ساختار Speaker در Source Metadata شامل سه سطح مهم است:
</p>

<ul>
<li><code>base</code>: شواهد اولیه Parser یا Source marker.</li>
<li><code>effective</code>: Speaker مؤثر حاصل از Attribution logic.</li>
<li><code>human_attribution_overlay</code>: اصلاح انسانی در صورت وجود.</li>
</ul>

<p>
بنابراین در این مرحله
<code>attribution.speaker</code>
با
<code>source_metadata.speaker.effective.canonical</code>
و
<code>effective.label</code>
مقایسه می‌شود.
</p>

<h4>هدف Discovery</h4>

<ul>
<li>بررسی اینکه 11,210 Attribution speaker موجود واقعاً با Canonical Speaker در Source Metadata سازگارند یا خیر.</li>
<li>تفکیک 5,016 Attribution speaker خالی به گروه‌های دارای Speaker label، Canonical speaker یا Unresolved speaker.</li>
<li>اندازه‌گیری وضعیت‌های <code>base.status</code> و <code>boundary_type</code>.</li>
<li>بررسی توزیع Speaker evidence به تفکیک Program.</li>
<li>تشخیص اینکه Null بودن Attribution speaker الزاماً Metadata loss است یا بازتاب سیاست Canonical attribution.</li>
</ul>

<h4>طبقه‌بندی Evidence</h4>

<p>
طبقه‌بندی‌های این مرحله فقط توصیفی هستند:
</p>

<ul>
<li><code>ATTRIBUTION_CANONICAL_MATCH</code>: Attribution با Effective canonical برابر است.</li>
<li><code>ATTRIBUTION_PRESENT_CANONICAL_MISMATCH</code>: هر دو Canonical هستند ولی مقدار متفاوت دارند.</li>
<li><code>ATTRIBUTION_PRESENT_NO_EFFECTIVE_CANONICAL</code>: Attribution وجود دارد ولی Canonical متناظر در Metadata وجود ندارد.</li>
<li><code>ATTRIBUTION_NULL_EFFECTIVE_CANONICAL_PRESENT</code>: Metadata Canonical دارد ولی Attribution خالی است.</li>
<li><code>ATTRIBUTION_NULL_LABEL_ONLY</code>: Canonical وجود ندارد ولی Speaker label واقعی وجود دارد.</li>
<li><code>ATTRIBUTION_NULL_UNRESOLVED</code>: نه Canonical و نه Label مؤثر وجود دارد.</li>
</ul>

<h4>اصل مهم</h4>

<p>
وجود Speaker label غیرCanonical مانند «مجری»، «خانم متولیان» یا «شرکت‌کننده 1»
نباید به‌طور خودکار به Canonical شخص دیگری تبدیل شود.
همچنین Unresolved speaker بدون Evidence مستقل نباید حدس زده شود.
</p>

<h4>معیار PASS</h4>

<p>
تمام 16,226 Transcript باید دقیقاً در یکی از Evidence classها قرار گیرند،
ساختار Speaker باید برای همه Recordها قابل خواندن باشد،
و تمام Attribution speakerهای Canonical موجود باید از نظر سازگاری با Metadata به‌صورت مستقل بررسی شوند.
</p>

</div>

```python
# Code Cell 24

from collections import Counter, defaultdict
import json

# -------------------------------------------------------------------
# Independent transcript population
# -------------------------------------------------------------------

speaker_audit_records = [
    record
    for record in quality_records
    if record["source_type"] == "transcript"
]

# -------------------------------------------------------------------
# Speaker structure parser
# -------------------------------------------------------------------

def parse_speaker_structure(record):
    speaker_object = (
        record.get(
            "source_metadata",
            {}
        ).get(
            "speaker"
        )
    )

    if not isinstance(
        speaker_object,
        dict,
    ):
        return {
            "structure_valid": False,
            "base_status": None,
            "boundary_type": None,
            "base_canonical": None,
            "base_raw": None,
            "effective_basis": None,
            "effective_canonical": None,
            "effective_label": None,
            "primary_dr_roudi_corpus": None,
            "human_overlay": None,
        }

    base = speaker_object.get(
        "base"
    )

    effective = speaker_object.get(
        "effective"
    )

    base = (
        base
        if isinstance(
            base,
            dict,
        )
        else {}
    )

    effective = (
        effective
        if isinstance(
            effective,
            dict,
        )
        else {}
    )

    return {
        "structure_valid": True,
        "base_status": base.get(
            "status"
        ),
        "boundary_type": base.get(
            "boundary_type"
        ),
        "base_canonical": base.get(
            "canonical"
        ),
        "base_raw": base.get(
            "raw"
        ),
        "effective_basis": effective.get(
            "basis"
        ),
        "effective_canonical": effective.get(
            "canonical"
        ),
        "effective_label": effective.get(
            "label"
        ),
        "primary_dr_roudi_corpus": effective.get(
            "primary_dr_roudi_corpus"
        ),
        "human_overlay": speaker_object.get(
            "human_attribution_overlay"
        ),
    }


def classify_speaker_evidence(
    attribution_speaker,
    parsed,
):
    effective_canonical = parsed[
        "effective_canonical"
    ]

    effective_label = parsed[
        "effective_label"
    ]

    if attribution_speaker is not None:
        if (
            effective_canonical
            == attribution_speaker
        ):
            return (
                "ATTRIBUTION_CANONICAL_MATCH"
            )

        if effective_canonical is not None:
            return (
                "ATTRIBUTION_PRESENT_CANONICAL_MISMATCH"
            )

        return (
            "ATTRIBUTION_PRESENT_NO_EFFECTIVE_CANONICAL"
        )

    if effective_canonical is not None:
        return (
            "ATTRIBUTION_NULL_EFFECTIVE_CANONICAL_PRESENT"
        )

    if (
        isinstance(
            effective_label,
            str,
        )
        and effective_label.strip()
    ):
        return (
            "ATTRIBUTION_NULL_LABEL_ONLY"
        )

    return (
        "ATTRIBUTION_NULL_UNRESOLVED"
    )


# -------------------------------------------------------------------
# Full speaker semantic audit
# -------------------------------------------------------------------

speaker_semantic_rows = []

for record in speaker_audit_records:
    parsed = parse_speaker_structure(
        record
    )

    attribution_speaker = (
        record.get(
            "attribution",
            {}
        ).get(
            "speaker"
        )
    )

    evidence_class = (
        classify_speaker_evidence(
            attribution_speaker,
            parsed,
        )
    )

    speaker_semantic_rows.append(
        {
            "document_id": record[
                "document_id"
            ],
            "source_id": record[
                "source_id"
            ],
            "program_id": record[
                "source_metadata"
            ].get(
                "program_id"
            ),
            "text": record[
                "text"
            ],
            "attribution_speaker": (
                attribution_speaker
            ),
            "evidence_class": (
                evidence_class
            ),
            **parsed,
        }
    )

# -------------------------------------------------------------------
# Global evidence classes
# -------------------------------------------------------------------

evidence_class_counts = Counter(
    row["evidence_class"]
    for row in speaker_semantic_rows
)

print(
    "=== TRANSCRIPT SPEAKER SEMANTIC AUDIT ==="
)

print(
    "Transcript records:",
    len(
        speaker_semantic_rows
    ),
)

print(
    "\nEvidence classes:"
)

evidence_class_order = [
    "ATTRIBUTION_CANONICAL_MATCH",
    "ATTRIBUTION_PRESENT_CANONICAL_MISMATCH",
    "ATTRIBUTION_PRESENT_NO_EFFECTIVE_CANONICAL",
    "ATTRIBUTION_NULL_EFFECTIVE_CANONICAL_PRESENT",
    "ATTRIBUTION_NULL_LABEL_ONLY",
    "ATTRIBUTION_NULL_UNRESOLVED",
]

for evidence_class in (
    evidence_class_order
):
    print(
        "-",
        evidence_class,
        ":",
        evidence_class_counts.get(
            evidence_class,
            0,
        ),
    )

# -------------------------------------------------------------------
# Base status / boundary type profile
# -------------------------------------------------------------------

print(
    "\n=== SPEAKER BASE STATUS PROFILE ==="
)

base_status_counts = Counter(
    row["base_status"]
    for row in speaker_semantic_rows
)

for value, count in (
    base_status_counts.most_common()
):
    print(
        "-",
        repr(value),
        ":",
        count,
    )

print(
    "\n=== SPEAKER BOUNDARY TYPE PROFILE ==="
)

boundary_type_counts = Counter(
    row["boundary_type"]
    for row in speaker_semantic_rows
)

for value, count in (
    boundary_type_counts.most_common()
):
    print(
        "-",
        repr(value),
        ":",
        count,
    )

# -------------------------------------------------------------------
# Effective canonical / label coverage
# -------------------------------------------------------------------

effective_canonical_present = sum(
    row[
        "effective_canonical"
    ]
    is not None
    for row in speaker_semantic_rows
)

effective_label_present = sum(
    isinstance(
        row[
            "effective_label"
        ],
        str,
    )
    and row[
        "effective_label"
    ].strip() != ""
    for row in speaker_semantic_rows
)

effective_unresolved = sum(
    row[
        "effective_canonical"
    ]
    is None
    and (
        row[
            "effective_label"
        ]
        is None
        or (
            isinstance(
                row[
                    "effective_label"
                ],
                str,
            )
            and row[
                "effective_label"
            ].strip()
            == ""
        )
    )
    for row in speaker_semantic_rows
)

print(
    "\n=== EFFECTIVE SPEAKER COVERAGE ==="
)

print(
    "Effective canonical present:",
    effective_canonical_present,
)

print(
    "Effective label present:",
    effective_label_present,
)

print(
    "No effective canonical/label:",
    effective_unresolved,
)

# -------------------------------------------------------------------
# Canonical value distributions
# -------------------------------------------------------------------

print(
    "\n=== EFFECTIVE CANONICAL VALUES ==="
)

canonical_counts = Counter(
    row[
        "effective_canonical"
    ]
    for row in speaker_semantic_rows
)

for value, count in (
    canonical_counts.most_common(
        20
    )
):
    print(
        "-",
        repr(value),
        ":",
        count,
    )

print(
    "\n=== EFFECTIVE LABEL VALUES ==="
)

label_counts = Counter(
    row[
        "effective_label"
    ]
    for row in speaker_semantic_rows
)

for value, count in (
    label_counts.most_common(
        40
    )
):
    print(
        "-",
        repr(value),
        ":",
        count,
    )

# -------------------------------------------------------------------
# Attribution NULL population decomposition
# -------------------------------------------------------------------

attribution_null_rows = [
    row
    for row in speaker_semantic_rows
    if row[
        "attribution_speaker"
    ]
    is None
]

print(
    "\n=== ATTRIBUTION.SPEAKER NULL DECOMPOSITION ==="
)

print(
    "Total attribution speaker NULL:",
    len(
        attribution_null_rows
    ),
)

null_class_counts = Counter(
    row[
        "evidence_class"
    ]
    for row in attribution_null_rows
)

for value, count in (
    null_class_counts.most_common()
):
    print(
        "-",
        value,
        ":",
        count,
    )

print(
    "\nBase status within attribution NULL:"
)

null_base_status_counts = Counter(
    row[
        "base_status"
    ]
    for row in attribution_null_rows
)

for value, count in (
    null_base_status_counts.most_common()
):
    print(
        "-",
        repr(value),
        ":",
        count,
    )

# -------------------------------------------------------------------
# Program-level decomposition
# -------------------------------------------------------------------

print(
    "\n=== PROGRAM-LEVEL SPEAKER EVIDENCE ==="
)

program_evidence = defaultdict(
    Counter
)

for row in speaker_semantic_rows:
    program_evidence[
        row[
            "program_id"
        ]
    ][
        row[
            "evidence_class"
        ]
    ] += 1

for program_id in sorted(
    program_evidence,
    key=lambda value: (
        str(value)
    ),
):
    counts = (
        program_evidence[
            program_id
        ]
    )

    total = sum(
        counts.values()
    )

    print(
        "\nPROGRAM:",
        program_id,
    )

    print(
        " Total:",
        total,
    )

    for evidence_class in (
        evidence_class_order
    ):
        count = counts.get(
            evidence_class,
            0,
        )

        if count:
            print(
                " -",
                evidence_class,
                ":",
                count,
            )

# -------------------------------------------------------------------
# Representative examples by semantic class
# -------------------------------------------------------------------

print(
    "\n=== REPRESENTATIVE SPEAKER EVIDENCE EXAMPLES ==="
)

examples_by_class = defaultdict(
    list
)

for row in speaker_semantic_rows:
    evidence_class = row[
        "evidence_class"
    ]

    if len(
        examples_by_class[
            evidence_class
        ]
    ) < 12:
        examples_by_class[
            evidence_class
        ].append(
            row
        )

for evidence_class in (
    evidence_class_order
):
    examples = (
        examples_by_class.get(
            evidence_class,
            []
        )
    )

    if not examples:
        continue

    print(
        "\n"
        + "=" * 100
    )

    print(
        "CLASS:",
        evidence_class,
    )

    print(
        "Count:",
        evidence_class_counts[
            evidence_class
        ],
    )

    for row in examples:
        print(
            "\n- document_id:",
            row[
                "document_id"
            ],
        )

        print(
            "  source_id:",
            row[
                "source_id"
            ],
        )

        print(
            "  program_id:",
            row[
                "program_id"
            ],
        )

        print(
            "  attribution:",
            repr(
                row[
                    "attribution_speaker"
                ]
            ),
        )

        print(
            "  effective canonical:",
            repr(
                row[
                    "effective_canonical"
                ]
            ),
        )

        print(
            "  effective label:",
            repr(
                row[
                    "effective_label"
                ]
            ),
        )

        print(
            "  base status:",
            repr(
                row[
                    "base_status"
                ]
            ),
        )

        print(
            "  boundary type:",
            repr(
                row[
                    "boundary_type"
                ]
            ),
        )

        print(
            "  text:",
            row[
                "text"
            ]
            .replace(
                "\n",
                " ",
            )[:400],
        )

# -------------------------------------------------------------------
# Structural validation
# -------------------------------------------------------------------

all_speaker_structures_valid = all(
    row[
        "structure_valid"
    ]
    for row in speaker_semantic_rows
)

population_accounted = (
    sum(
        evidence_class_counts.values()
    )
    == len(
        speaker_semantic_rows
    )
)

attribution_present_count = sum(
    row[
        "attribution_speaker"
    ]
    is not None
    for row in speaker_semantic_rows
)

attribution_null_count = sum(
    row[
        "attribution_speaker"
    ]
    is None
    for row in speaker_semantic_rows
)

attribution_population_ok = (
    attribution_present_count
    == 11210
    and attribution_null_count
    == 5016
)

canonical_mismatch_count = (
    evidence_class_counts.get(
        "ATTRIBUTION_PRESENT_CANONICAL_MISMATCH",
        0,
    )
)

present_without_canonical_count = (
    evidence_class_counts.get(
        "ATTRIBUTION_PRESENT_NO_EFFECTIVE_CANONICAL",
        0,
    )
)

print(
    "\nValidation:"
)

print(
    "All source_metadata.speaker structures valid:",
    "PASS"
    if all_speaker_structures_valid
    else "FAIL",
)

print(
    "Speaker evidence population accounted:",
    "PASS"
    if population_accounted
    else "FAIL",
)

print(
    "Known attribution population reproduced:",
    "PASS"
    if attribution_population_ok
    else "FAIL",
)

print(
    "Canonical mismatches discovered:",
    canonical_mismatch_count,
)

print(
    "Attribution present without effective canonical:",
    present_without_canonical_count,
)

cell_gate = (
    all_speaker_structures_valid
    and population_accounted
    and attribution_population_ok
)

print(
    "CODE CELL 24 GATE:",
    "PASS"
    if cell_gate
    else "FAIL",
)
```

```python
# Code Cell 25

from collections import Counter, defaultdict
from pathlib import Path
import re

# -------------------------------------------------------------------
# Repository policy discovery
# -------------------------------------------------------------------

policy_search_terms = [
    "EVIDENCE_REQUIRED",
    "quarantine",
    "quarantined",
    "quality",
    "quality_score",
    "release",
    "release gate",
    "release_gate",
    "accepted dataset",
    "accepted_dataset",
    "rights",
    "rights_status",
    "rights evidence",
    "rights_evidence",
    "attribution",
    "speaker",
    "dedup",
    "duplicate",
    "near-duplicate",
    "near_duplicate",
    "canonical relation",
    "canonical_relation",
]

text_suffixes = {
    ".md",
    ".json",
    ".jsonl",
    ".txt",
    ".yaml",
    ".yml",
    ".py",
    ".ipynb",
}

excluded_parts = {
    ".git",
    ".ipynb_checkpoints",
    "__pycache__",
    ".venv",
    "venv",
    "node_modules",
}

policy_matches = defaultdict(
    list
)

scanned_files = 0
readable_files = 0

for path in sorted(
    repo_root.rglob("*")
):
    if not path.is_file():
        continue

    if any(
        part in excluded_parts
        for part in path.parts
    ):
        continue

    if path.suffix.lower() not in text_suffixes:
        continue

    scanned_files += 1

    try:
        text = path.read_text(
            encoding="utf-8"
        )
    except (
        UnicodeDecodeError,
        OSError,
    ):
        continue

    readable_files += 1

    relative_path = str(
        path.relative_to(
            repo_root
        )
    )

    lower_text = text.lower()

    for term in policy_search_terms:
        if term.lower() in lower_text:
            policy_matches[
                relative_path
            ].append(
                term
            )

# -------------------------------------------------------------------
# Candidate-file ranking
# -------------------------------------------------------------------

policy_candidate_files = []

for relative_path, terms in (
    policy_matches.items()
):
    unique_terms = sorted(
        set(terms)
    )

    path_lower = (
        relative_path.lower()
    )

    evidence_weight = 0

    for keyword in [
        "roadmap",
        "manifest",
        "acceptance",
        "report",
        "policy",
        "phase2",
        "notebook",
    ]:
        if keyword in path_lower:
            evidence_weight += 1

    policy_candidate_files.append(
        {
            "path": relative_path,
            "matched_terms": (
                unique_terms
            ),
            "term_count": len(
                unique_terms
            ),
            "evidence_weight": (
                evidence_weight
            ),
        }
    )

policy_candidate_files.sort(
    key=lambda row: (
        -row["evidence_weight"],
        -row["term_count"],
        row["path"],
    )
)

print(
    "=== REPOSITORY QUALITY/POLICY DISCOVERY ==="
)

print(
    "Text-like files scanned:",
    scanned_files,
)

print(
    "Readable files:",
    readable_files,
)

print(
    "Files with policy-related matches:",
    len(
        policy_candidate_files
    ),
)

print(
    "\n=== HIGHEST-SIGNAL POLICY FILES ==="
)

for row in policy_candidate_files[
    :60
]:
    print(
        "\n-",
        row["path"],
    )

    print(
        "  evidence weight:",
        row[
            "evidence_weight"
        ],
    )

    print(
        "  matched terms:",
        row[
            "matched_terms"
        ],
    )

# -------------------------------------------------------------------
# Term-level file coverage
# -------------------------------------------------------------------

print(
    "\n=== POLICY TERM COVERAGE ==="
)

term_file_counts = Counter()

for row in policy_candidate_files:
    for term in row[
        "matched_terms"
    ]:
        term_file_counts[
            term
        ] += 1

for term in policy_search_terms:
    print(
        "-",
        term,
        ":",
        term_file_counts.get(
            term,
            0,
        ),
        "files",
    )

# -------------------------------------------------------------------
# Context extraction
# -------------------------------------------------------------------

priority_terms = [
    "EVIDENCE_REQUIRED",
    "quarantine",
    "quarantined",
    "release gate",
    "release_gate",
    "rights",
    "rights evidence",
    "rights_evidence",
    "quality",
    "accepted dataset",
    "accepted_dataset",
    "attribution",
    "dedup",
    "near-duplicate",
    "near_duplicate",
]

# Prefer high-signal files while keeping output bounded.
context_candidate_paths = [
    repo_root
    / row["path"]
    for row in policy_candidate_files[
        :40
    ]
]

policy_context_hits = []

for path in context_candidate_paths:
    try:
        lines = path.read_text(
            encoding="utf-8"
        ).splitlines()
    except (
        UnicodeDecodeError,
        OSError,
    ):
        continue

    relative_path = str(
        path.relative_to(
            repo_root
        )
    )

    for line_index, line in enumerate(
        lines
    ):
        lower_line = (
            line.lower()
        )

        matched = [
            term
            for term in priority_terms
            if term.lower()
            in lower_line
        ]

        if not matched:
            continue

        start = max(
            0,
            line_index - 2,
        )

        end = min(
            len(lines),
            line_index + 3,
        )

        context = "\n".join(
            f"{idx + 1}: {lines[idx]}"
            for idx in range(
                start,
                end,
            )
        )

        policy_context_hits.append(
            {
                "path": (
                    relative_path
                ),
                "line": (
                    line_index + 1
                ),
                "matched_terms": sorted(
                    set(
                        matched
                    )
                ),
                "context": context,
            }
        )

# -------------------------------------------------------------------
# Deduplicate heavily overlapping contexts
# -------------------------------------------------------------------

deduplicated_context_hits = []

seen_context_keys = set()

for hit in policy_context_hits:
    normalized_context = re.sub(
        r"\s+",
        " ",
        hit["context"],
    ).strip()

    key = (
        hit["path"],
        normalized_context,
    )

    if key in seen_context_keys:
        continue

    seen_context_keys.add(
        key
    )

    deduplicated_context_hits.append(
        hit
    )

print(
    "\n=== POLICY CONTEXT HITS ==="
)

print(
    "Context hits:",
    len(
        deduplicated_context_hits
    ),
)

for index, hit in enumerate(
    deduplicated_context_hits[
        :100
    ],
    start=1,
):
    print(
        "\n"
        + "=" * 100
    )

    print(
        "HIT",
        index,
    )

    print(
        "File:",
        hit["path"],
    )

    print(
        "Anchor line:",
        hit["line"],
    )

    print(
        "Terms:",
        hit[
            "matched_terms"
        ],
    )

    print(
        hit["context"]
    )

# -------------------------------------------------------------------
# Focused EVIDENCE_REQUIRED discovery
# -------------------------------------------------------------------

print(
    "\n=== EVIDENCE_REQUIRED FOCUSED DISCOVERY ==="
)

evidence_required_hits = []

for row in policy_candidate_files:
    path = (
        repo_root
        / row["path"]
    )

    try:
        lines = path.read_text(
            encoding="utf-8"
        ).splitlines()
    except (
        UnicodeDecodeError,
        OSError,
    ):
        continue

    for line_index, line in enumerate(
        lines
    ):
        if (
            "EVIDENCE_REQUIRED"
            not in line
        ):
            continue

        start = max(
            0,
            line_index - 3,
        )

        end = min(
            len(lines),
            line_index + 4,
        )

        evidence_required_hits.append(
            {
                "path": row[
                    "path"
                ],
                "line": (
                    line_index + 1
                ),
                "context": "\n".join(
                    f"{idx + 1}: {lines[idx]}"
                    for idx in range(
                        start,
                        end,
                    )
                ),
            }
        )

print(
    "EVIDENCE_REQUIRED occurrences:",
    len(
        evidence_required_hits
    ),
)

for index, hit in enumerate(
    evidence_required_hits[
        :60
    ],
    start=1,
):
    print(
        "\n"
        + "-" * 100
    )

    print(
        "EVIDENCE_REQUIRED HIT",
        index,
    )

    print(
        "File:",
        hit["path"],
    )

    print(
        "Anchor line:",
        hit["line"],
    )

    print(
        hit["context"]
    )

# -------------------------------------------------------------------
# Validation
# -------------------------------------------------------------------

search_executed = (
    scanned_files > 0
    and readable_files > 0
)

policy_files_found = (
    len(
        policy_candidate_files
    )
    > 0
)

context_found = (
    len(
        deduplicated_context_hits
    )
    > 0
)

evidence_required_discovery_completed = (
    isinstance(
        evidence_required_hits,
        list,
    )
)

print(
    "\nValidation:"
)

print(
    "Repository text search executed:",
    "PASS"
    if search_executed
    else "FAIL",
)

print(
    "Policy-related files discovered:",
    "PASS"
    if policy_files_found
    else "FAIL",
)

print(
    "Policy context extracted:",
    "PASS"
    if context_found
    else "FAIL",
)

print(
    "EVIDENCE_REQUIRED focused discovery completed:",
    "PASS"
    if evidence_required_discovery_completed
    else "FAIL",
)

cell_gate = (
    search_executed
    and policy_files_found
    and context_found
    and evidence_required_discovery_completed
)

print(
    "CODE CELL 25 GATE:",
    "PASS"
    if cell_gate
    else "FAIL",
)
```

```python
# Code Cell 26

from collections import Counter
from pathlib import Path
import json
import re

# -------------------------------------------------------------------
# Previously discovered repository artifacts
# -------------------------------------------------------------------

speaker_alias_path = (
    repo_root
    / "data/phase2/interim/speaker_aliases_v1.json"
)

transcript_manifest_path = (
    repo_root
    / "data/phase2/manifests/transcript_manifest_v1.json"
)

book_manifest_path = (
    repo_root
    / "data/phase2/manifests/book_manifest_v1.json"
)

website_acceptance_path = (
    repo_root
    / "reports/phase2/website_acceptance_v1.json"
)

required_policy_paths = [
    speaker_alias_path,
    transcript_manifest_path,
    book_manifest_path,
    website_acceptance_path,
]

print(
    "=== AUTHORITY CONTRACT ARTIFACT DISCOVERY ==="
)

for path in required_policy_paths:
    print(
        "-",
        path.relative_to(
            repo_root
        ),
        "| exists:",
        path.exists(),
    )

# -------------------------------------------------------------------
# Load artifacts
# -------------------------------------------------------------------

with speaker_alias_path.open(
    "r",
    encoding="utf-8",
) as handle:
    speaker_alias_data = json.load(
        handle
    )

with transcript_manifest_path.open(
    "r",
    encoding="utf-8",
) as handle:
    transcript_manifest_data = json.load(
        handle
    )

with book_manifest_path.open(
    "r",
    encoding="utf-8",
) as handle:
    book_manifest_data = json.load(
        handle
    )

with website_acceptance_path.open(
    "r",
    encoding="utf-8",
) as handle:
    website_acceptance_data = json.load(
        handle
    )

print(
    "\nTop-level structures:"
)

for name, data in [
    (
        "speaker_aliases",
        speaker_alias_data,
    ),
    (
        "transcript_manifest",
        transcript_manifest_data,
    ),
    (
        "book_manifest",
        book_manifest_data,
    ),
    (
        "website_acceptance",
        website_acceptance_data,
    ),
]:
    if isinstance(
        data,
        dict,
    ):
        top_level = sorted(
            data.keys()
        )
    else:
        top_level = (
            f"{type(data).__name__}"
        )

    print(
        "-",
        name,
        ":",
        top_level,
    )

# -------------------------------------------------------------------
# Generic recursive scalar walker
# -------------------------------------------------------------------

def walk_scalars(
    value,
    path="$",
):
    if isinstance(
        value,
        dict,
    ):
        for key in sorted(
            value.keys(),
            key=str,
        ):
            yield from walk_scalars(
                value[key],
                (
                    f"{path}."
                    f"{key}"
                ),
            )

    elif isinstance(
        value,
        list,
    ):
        for index, item in enumerate(
            value
        ):
            yield from walk_scalars(
                item,
                (
                    f"{path}"
                    f"[{index}]"
                ),
            )

    else:
        yield (
            path,
            value,
        )

# -------------------------------------------------------------------
# Speaker alias evidence
# -------------------------------------------------------------------

print(
    "\n=== SPEAKER ALIAS EVIDENCE ==="
)

alias_scalar_rows = list(
    walk_scalars(
        speaker_alias_data
    )
)

print(
    "Scalar values in alias artifact:",
    len(
        alias_scalar_rows
    ),
)

speaker_interest_patterns = [
    "roudi",
    "رودی",
    "کمیل",
    "speaker",
    "alias",
    "canonical",
    "استاد",
]

alias_evidence_rows = []

for scalar_path, value in (
    alias_scalar_rows
):
    searchable = (
        f"{scalar_path} "
        f"{value}"
    ).lower()

    if any(
        pattern.lower()
        in searchable
        for pattern
        in speaker_interest_patterns
    ):
        alias_evidence_rows.append(
            (
                scalar_path,
                value,
            )
        )

print(
    "Speaker/alias evidence rows:",
    len(
        alias_evidence_rows
    ),
)

for scalar_path, value in (
    alias_evidence_rows[:300]
):
    print(
        "-",
        scalar_path,
    )
    print(
        " ",
        repr(value),
    )

# -------------------------------------------------------------------
# Exact values potentially referring to Dr. Komeil Roudi
# Discovery only — no new alias is inferred.
# -------------------------------------------------------------------

roudi_related_values = Counter()

for scalar_path, value in (
    alias_scalar_rows
):
    if not isinstance(
        value,
        str,
    ):
        continue

    normalized = (
        value.strip()
    )

    if (
        "roudi"
        in normalized.lower()
        or "رودی"
        in normalized
        or "کمیل"
        in normalized
    ):
        roudi_related_values[
            normalized
        ] += 1

print(
    "\n=== EXPLICIT ROUDI-RELATED VALUES IN ALIAS ARTIFACT ==="
)

for value, count in (
    roudi_related_values.most_common()
):
    print(
        "-",
        repr(value),
        ":",
        count,
    )

# -------------------------------------------------------------------
# Transcript manifest policy evidence
# -------------------------------------------------------------------

print(
    "\n=== TRANSCRIPT MANIFEST SPEAKER/POLICY EVIDENCE ==="
)

manifest_scalar_rows = list(
    walk_scalars(
        transcript_manifest_data
    )
)

manifest_policy_patterns = [
    "speaker",
    "attribution",
    "unknown",
    "question",
    "response",
    "primary",
    "roudi",
    "رودی",
    "کمیل",
    "rights",
    "invent",
]

manifest_policy_rows = []

for scalar_path, value in (
    manifest_scalar_rows
):
    searchable = (
        f"{scalar_path} "
        f"{value}"
    ).lower()

    if any(
        pattern.lower()
        in searchable
        for pattern
        in manifest_policy_patterns
    ):
        manifest_policy_rows.append(
            (
                scalar_path,
                value,
            )
        )

print(
    "Relevant manifest scalar rows:",
    len(
        manifest_policy_rows
    ),
)

for scalar_path, value in (
    manifest_policy_rows[:350]
):
    print(
        "-",
        scalar_path,
    )
    print(
        " ",
        repr(value),
    )

# -------------------------------------------------------------------
# Reproduce user-confirmed authority populations from current dataset
# -------------------------------------------------------------------

print(
    "\n=== CURRENT DATASET AUTHORITY POPULATION ==="
)

website_records_current = [
    record
    for record in quality_records
    if record["source_type"]
    == "website"
]

book_records_current = [
    record
    for record in quality_records
    if record["source_type"]
    == "book"
]

transcript_records_current = [
    record
    for record in quality_records
    if record["source_type"]
    == "transcript"
]

website_origin_counts = Counter(
    record.get(
        "content_origin"
    )
    for record in website_records_current
)

website_knowledge_owner_counts = Counter(
    record.get(
        "attribution",
        {},
    ).get(
        "knowledge_owner"
    )
    for record
    in website_records_current
)

book_author_counts = Counter(
    record.get(
        "attribution",
        {},
    ).get(
        "author"
    )
    for record
    in book_records_current
)

book_knowledge_owner_counts = Counter(
    record.get(
        "attribution",
        {},
    ).get(
        "knowledge_owner"
    )
    for record
    in book_records_current
)

transcript_attribution_speaker_counts = Counter(
    record.get(
        "attribution",
        {},
    ).get(
        "speaker"
    )
    for record
    in transcript_records_current
)

print(
    "Website content_origin:",
    dict(
        website_origin_counts
    ),
)

print(
    "Website knowledge_owner:",
    dict(
        website_knowledge_owner_counts
    ),
)

print(
    "Book author:",
    dict(
        book_author_counts
    ),
)

print(
    "Book knowledge_owner:",
    dict(
        book_knowledge_owner_counts
    ),
)

print(
    "Transcript attribution.speaker:",
    dict(
        transcript_attribution_speaker_counts
    ),
)

# -------------------------------------------------------------------
# Reuse semantic speaker audit from Code Cell 24
# -------------------------------------------------------------------

speaker_evidence_counts_current = Counter(
    row[
        "evidence_class"
    ]
    for row
    in speaker_semantic_rows
)

print(
    "\nTranscript semantic speaker evidence:"
)

for key, value in sorted(
    speaker_evidence_counts_current.items()
):
    print(
        "-",
        key,
        ":",
        value,
    )

# -------------------------------------------------------------------
# Existing Website acceptance authority contract
# -------------------------------------------------------------------

print(
    "\n=== WEBSITE ACCEPTANCE AUTHORITY EVIDENCE ==="
)

website_acceptance_scalars = list(
    walk_scalars(
        website_acceptance_data
    )
)

website_authority_rows = []

for scalar_path, value in (
    website_acceptance_scalars
):
    searchable = (
        f"{scalar_path} "
        f"{value}"
    ).lower()

    if any(
        term
        in searchable
        for term in [
            "primary_author",
            "supervised",
            "knowledge_owner",
            "attribution",
        ]
    ):
        website_authority_rows.append(
            (
                scalar_path,
                value,
            )
        )

for scalar_path, value in (
    website_authority_rows[:120]
):
    print(
        "-",
        scalar_path,
        "=",
        repr(value),
    )

# -------------------------------------------------------------------
# Existing Book authority evidence
# -------------------------------------------------------------------

print(
    "\n=== BOOK MANIFEST AUTHORITY EVIDENCE ==="
)

book_manifest_scalars = list(
    walk_scalars(
        book_manifest_data
    )
)

book_authority_rows = []

for scalar_path, value in (
    book_manifest_scalars
):
    searchable = (
        f"{scalar_path} "
        f"{value}"
    ).lower()

    if any(
        term
        in searchable
        for term in [
            "author",
            "knowledge_owner",
            "authorship",
            "roudi",
            "رودی",
            "کمیل",
        ]
    ):
        book_authority_rows.append(
            (
                scalar_path,
                value,
            )
        )

for scalar_path, value in (
    book_authority_rows[:160]
):
    print(
        "-",
        scalar_path,
        "=",
        repr(value),
    )

# -------------------------------------------------------------------
# Validation
# -------------------------------------------------------------------

required_files_present = all(
    path.exists()
    for path
    in required_policy_paths
)

speaker_alias_loaded = (
    speaker_alias_data
    is not None
)

manifest_policy_discovered = (
    len(
        manifest_policy_rows
    )
    > 0
)

website_authority_reproduced = (
    website_origin_counts
    == Counter(
        {
            "primary_author": 76,
            "supervised": 453,
        }
    )
    and website_knowledge_owner_counts
    == Counter(
        {
            "Dr. Komeil Roudi": 529,
        }
    )
)

book_authority_reproduced = (
    book_author_counts
    == Counter(
        {
            "Dr. Komeil Roudi": 2701,
        }
    )
    and book_knowledge_owner_counts
    == Counter(
        {
            "Dr. Komeil Roudi": 2701,
        }
    )
)

transcript_known_authority_reproduced = (
    transcript_attribution_speaker_counts.get(
        "Dr. Komeil Roudi",
        0,
    )
    == 11210
    and transcript_attribution_speaker_counts.get(
        None,
        0,
    )
    == 5016
)

speaker_semantic_population_ok = (
    sum(
        speaker_evidence_counts_current.values()
    )
    == 16226
)

print(
    "\nValidation:"
)

print(
    "Required policy artifacts present:",
    "PASS"
    if required_files_present
    else "FAIL",
)

print(
    "Speaker alias artifact loaded:",
    "PASS"
    if speaker_alias_loaded
    else "FAIL",
)

print(
    "Transcript speaker/policy evidence discovered:",
    "PASS"
    if manifest_policy_discovered
    else "FAIL",
)

print(
    "Website authority population reproduced:",
    "PASS"
    if website_authority_reproduced
    else "FAIL",
)

print(
    "Book authority population reproduced:",
    "PASS"
    if book_authority_reproduced
    else "FAIL",
)

print(
    "Transcript canonical authority population reproduced:",
    "PASS"
    if transcript_known_authority_reproduced
    else "FAIL",
)

print(
    "Transcript semantic speaker population preserved:",
    "PASS"
    if speaker_semantic_population_ok
    else "FAIL",
)

cell_gate = (
    required_files_present
    and speaker_alias_loaded
    and manifest_policy_discovered
    and website_authority_reproduced
    and book_authority_reproduced
    and transcript_known_authority_reproduced
    and speaker_semantic_population_ok
)

print(
    "CODE CELL 26 GATE:",
    "PASS"
    if cell_gate
    else "FAIL",
)
```

در مرحله قبل مشخص شد که
<code>Dr. Komeil Roudi</code>
Primary Speaker و Primary Subject پروژه است،
اما در Transcriptهای چندنفره Attribution باید فقط بر اساس Evidence واقعی انجام شود.
</p>

<p>
در Dataset فعلی تعداد
<code>11,210</code>
Segment مستقیماً دارای
<code>attribution.speaker = Dr. Komeil Roudi</code>
هستند و
<code>5,016</code>
Segment Attribution canonical ندارند.
</p>

<p>
از میان این 5,016 Segment، بعضی دارای
<code>effective.label</code>
هستند.
اما Label به‌تنهایی مجوز تبدیل به Canonical Speaker نیست.
تنها زمانی می‌توان یک Label را Evidence برای Primary Speaker دانست که همان Program در
<code>speaker_aliases_v1.json</code>
صراحتاً آن Alias را به
<code>Dr. Komeil Roudi</code>
نگاشت کرده باشد.
</p>

<h4>قواعد این مرحله</h4>

<ul>
<li>Canonical Attribution موجود بدون تغییر معتبر باقی می‌ماند.</li>
<li>Alias فقط در محدوده همان Program معتبر است.</li>
<li>هیچ Alias سراسری جدیدی ساخته نمی‌شود.</li>
<li>Labelهایی مانند «مجری»، نام مهمان، شرکت‌کننده یا نقش عمومی Primary تلقی نمی‌شوند مگر Registry صریحاً چنین رابطه‌ای ثبت کرده باشد.</li>
<li>Unresolved speaker همچنان Unresolved باقی می‌ماند.</li>
<li>در این مرحله هیچ فیلدی در Dataset تغییر نمی‌کند.</li>
</ul>

<h4>Evidence Classهای توصیفی</h4>

<ul>
<li><code>PRIMARY_CANONICAL_ATTRIBUTION</code>: Attribution فعلی مستقیماً دکتر کمیل رودی است.</li>
<li><code>PRIMARY_EXPLICIT_PROGRAM_ALIAS</code>: Attribution خالی است ولی Label دقیقاً در Alias Registry همان Program به دکتر نگاشت شده است.</li>
<li><code>NONPRIMARY_EXPLICIT_CONTEXT</code>: Speaker label وجود دارد اما Registry آن را Primary معرفی نمی‌کند.</li>
<li><code>UNRESOLVED_SPEAKER_CONTEXT</code>: Speaker هنوز بدون Canonical یا Label معتبر است.</li>
</ul>

<p>
این Classification در ادامه می‌تواند پایه Authority Policy شود،
اما هنوز Weight عددی، Retrieval Boost یا Quality Score ایجاد نمی‌شود.
</p>

<h4>معیار PASS</h4>

<p>
تمام 16,226 Transcript باید دقیقاً در یکی از چهار Class قرار گیرند،
تمام Aliasهای استفاده‌شده باید مستقیماً از Registry همان Program آمده باشند،
هیچ Canonical Attribution موجود نباید با Registry در تعارض قرار گیرد
و تعداد کل Population باید بدون تغییر باقی بماند.
</p>

```python
# Code Cell 27

from collections import Counter, defaultdict
import json

PRIMARY_CANONICAL_SPEAKER = (
    speaker_alias_data[
        "canonical_primary_speaker"
    ]
)

print(
    "=== TRANSCRIPT PRIMARY AUTHORITY RESOLUTION ==="
)

print(
    "Canonical primary speaker:",
    PRIMARY_CANONICAL_SPEAKER,
)

# -------------------------------------------------------------------
# Build program-scoped alias registry
# -------------------------------------------------------------------

program_alias_registry = {}

program_policy_registry = {}

program_entries = (
    speaker_alias_data.get(
        "programs",
        []
    )
)

for program_entry in program_entries:
    if not isinstance(
        program_entry,
        dict,
    ):
        continue

    # Discover program identifier from the actual artifact.
    program_id = None

    for candidate_key in [
        "program_id",
        "id",
        "program",
    ]:
        candidate_value = (
            program_entry.get(
                candidate_key
            )
        )

        if isinstance(
            candidate_value,
            str,
        ) and candidate_value.strip():
            program_id = (
                candidate_value
            )
            break

    if program_id is None:
        continue

    raw_aliases = (
        program_entry.get(
            "speaker_aliases",
            {}
        )
    )

    if not isinstance(
        raw_aliases,
        dict,
    ):
        raw_aliases = {}

    program_alias_registry[
        program_id
    ] = dict(
        raw_aliases
    )

    policy = (
        program_entry.get(
            "speaker_policy",
            {}
        )
    )

    program_policy_registry[
        program_id
    ] = (
        policy
        if isinstance(
            policy,
            dict,
        )
        else {}
    )

print(
    "Programs in alias registry:",
    len(
        program_alias_registry
    ),
)

print(
    "\n=== PROGRAM ALIASES MAPPING TO PRIMARY ==="
)

primary_aliases_by_program = {}

for program_id in sorted(
    program_alias_registry
):
    aliases = (
        program_alias_registry[
            program_id
        ]
    )

    primary_aliases = {
        alias: canonical
        for alias, canonical
        in aliases.items()
        if canonical
        == PRIMARY_CANONICAL_SPEAKER
    }

    primary_aliases_by_program[
        program_id
    ] = primary_aliases

    if primary_aliases:
        print(
            "\nPROGRAM:",
            program_id,
        )

        for alias, canonical in sorted(
            primary_aliases.items()
        ):
            print(
                "-",
                repr(alias),
                "->",
                repr(canonical),
            )

# -------------------------------------------------------------------
# Human attribution overlays from registry
# Discovery only.
# -------------------------------------------------------------------

human_overlay_records = (
    speaker_alias_data.get(
        "human_attribution_overlays",
        []
    )
)

print(
    "\n=== HUMAN ATTRIBUTION OVERLAY PROFILE ==="
)

print(
    "Overlay records:",
    len(
        human_overlay_records
    ),
)

overlay_role_counts = Counter()
overlay_primary_counts = Counter()

for overlay in human_overlay_records:
    if not isinstance(
        overlay,
        dict,
    ):
        continue

    overlay_role_counts[
        overlay.get(
            "speaker_role_for_project"
        )
    ] += 1

    overlay_primary_counts[
        overlay.get(
            "primary_dr_roudi_corpus"
        )
    ] += 1

print(
    "Speaker roles:",
    dict(
        overlay_role_counts
    ),
)

print(
    "Primary Dr. Roudi corpus flags:",
    dict(
        overlay_primary_counts
    ),
)

# -------------------------------------------------------------------
# Authority classifier
# -------------------------------------------------------------------

def classify_transcript_authority(
    row,
):
    attribution_speaker = (
        row[
            "attribution_speaker"
        ]
    )

    program_id = row[
        "program_id"
    ]

    effective_label = row[
        "effective_label"
    ]

    effective_canonical = row[
        "effective_canonical"
    ]

    # Existing canonical attribution is strongest.
    if (
        attribution_speaker
        == PRIMARY_CANONICAL_SPEAKER
    ):
        return {
            "authority_class": (
                "PRIMARY_CANONICAL_ATTRIBUTION"
            ),
            "authority_basis": (
                "CURRENT_DOCUMENT_ATTRIBUTION"
            ),
            "resolved_primary_alias": None,
        }

    # Any other existing canonical speaker
    # must never be overwritten by an alias rule.
    if attribution_speaker is not None:
        return {
            "authority_class": (
                "NONPRIMARY_EXPLICIT_CONTEXT"
            ),
            "authority_basis": (
                "CURRENT_DOCUMENT_NONPRIMARY_ATTRIBUTION"
            ),
            "resolved_primary_alias": None,
        }

    # If source metadata already has a canonical
    # identity but the top-level attribution does not,
    # do not silently promote it here.
    if effective_canonical is not None:
        return {
            "authority_class": (
                "NONPRIMARY_EXPLICIT_CONTEXT"
            ),
            "authority_basis": (
                "EFFECTIVE_CANONICAL_PRESENT_WITHOUT_DOCUMENT_ATTRIBUTION"
            ),
            "resolved_primary_alias": None,
        }

    program_aliases = (
        primary_aliases_by_program.get(
            program_id,
            {},
        )
    )

    if (
        isinstance(
            effective_label,
            str,
        )
        and effective_label
        in program_aliases
    ):
        return {
            "authority_class": (
                "PRIMARY_EXPLICIT_PROGRAM_ALIAS"
            ),
            "authority_basis": (
                "PROGRAM_SCOPED_ALIAS_REGISTRY"
            ),
            "resolved_primary_alias": (
                effective_label
            ),
        }

    if (
        isinstance(
            effective_label,
            str,
        )
        and effective_label.strip()
    ):
        return {
            "authority_class": (
                "NONPRIMARY_EXPLICIT_CONTEXT"
            ),
            "authority_basis": (
                "SOURCE_OBSERVED_LABEL_NOT_PRIMARY_ALIAS"
            ),
            "resolved_primary_alias": None,
        }

    return {
        "authority_class": (
            "UNRESOLVED_SPEAKER_CONTEXT"
        ),
        "authority_basis": (
            "NO_CANONICAL_OR_PRIMARY_ALIAS_EVIDENCE"
        ),
        "resolved_primary_alias": None,
    }

# -------------------------------------------------------------------
# Apply descriptive authority classification
# -------------------------------------------------------------------

transcript_authority_rows = []

for row in speaker_semantic_rows:
    authority = (
        classify_transcript_authority(
            row
        )
    )

    transcript_authority_rows.append(
        {
            **row,
            **authority,
        }
    )

authority_class_counts = Counter(
    row[
        "authority_class"
    ]
    for row
    in transcript_authority_rows
)

authority_basis_counts = Counter(
    row[
        "authority_basis"
    ]
    for row
    in transcript_authority_rows
)

print(
    "\n=== AUTHORITY CLASS DISTRIBUTION ==="
)

authority_class_order = [
    "PRIMARY_CANONICAL_ATTRIBUTION",
    "PRIMARY_EXPLICIT_PROGRAM_ALIAS",
    "NONPRIMARY_EXPLICIT_CONTEXT",
    "UNRESOLVED_SPEAKER_CONTEXT",
]

for authority_class in (
    authority_class_order
):
    print(
        "-",
        authority_class,
        ":",
        authority_class_counts.get(
            authority_class,
            0,
        ),
    )

print(
    "\n=== AUTHORITY BASIS DISTRIBUTION ==="
)

for basis, count in (
    authority_basis_counts.most_common()
):
    print(
        "-",
        basis,
        ":",
        count,
    )

# -------------------------------------------------------------------
# Alias-derived primary population
# -------------------------------------------------------------------

alias_primary_rows = [
    row
    for row
    in transcript_authority_rows
    if row[
        "authority_class"
    ]
    == "PRIMARY_EXPLICIT_PROGRAM_ALIAS"
]

print(
    "\n=== PRIMARY EXPLICIT PROGRAM-ALIAS POPULATION ==="
)

print(
    "Alias-derived primary segments:",
    len(
        alias_primary_rows
    ),
)

alias_primary_program_counts = Counter(
    row[
        "program_id"
    ]
    for row
    in alias_primary_rows
)

alias_primary_label_counts = Counter(
    (
        row[
            "program_id"
        ],
        row[
            "resolved_primary_alias"
        ],
    )
    for row
    in alias_primary_rows
)

print(
    "\nBy program:"
)

for program_id, count in (
    alias_primary_program_counts.most_common()
):
    print(
        "-",
        program_id,
        ":",
        count,
    )

print(
    "\nBy explicit program alias:"
)

for (
    program_id,
    alias,
), count in (
    alias_primary_label_counts.most_common()
):
    print(
        "-",
        program_id,
        "|",
        repr(alias),
        ":",
        count,
    )

# -------------------------------------------------------------------
# Non-primary explicit context profile
# -------------------------------------------------------------------

nonprimary_rows = [
    row
    for row
    in transcript_authority_rows
    if row[
        "authority_class"
    ]
    == "NONPRIMARY_EXPLICIT_CONTEXT"
]

nonprimary_label_counts = Counter(
    (
        row[
            "program_id"
        ],
        row[
            "effective_label"
        ],
    )
    for row
    in nonprimary_rows
)

print(
    "\n=== NONPRIMARY EXPLICIT CONTEXT ==="
)

print(
    "Segments:",
    len(
        nonprimary_rows
    ),
)

print(
    "Top program/label pairs:"
)

for (
    program_id,
    label,
), count in (
    nonprimary_label_counts.most_common(
        60
    )
):
    print(
        "-",
        program_id,
        "|",
        repr(label),
        ":",
        count,
    )

# -------------------------------------------------------------------
# Unresolved population by program
# -------------------------------------------------------------------

unresolved_authority_rows = [
    row
    for row
    in transcript_authority_rows
    if row[
        "authority_class"
    ]
    == "UNRESOLVED_SPEAKER_CONTEXT"
]

unresolved_by_program = Counter(
    row[
        "program_id"
    ]
    for row
    in unresolved_authority_rows
)

print(
    "\n=== UNRESOLVED SPEAKER CONTEXT BY PROGRAM ==="
)

print(
    "Segments:",
    len(
        unresolved_authority_rows
    ),
)

for program_id, count in (
    unresolved_by_program.most_common()
):
    print(
        "-",
        program_id,
        ":",
        count,
    )

# -------------------------------------------------------------------
# Program policy context for alias-derived primary rows
# -------------------------------------------------------------------

print(
    "\n=== PROGRAM POLICY FOR ALIAS-DERIVED PRIMARY ==="
)

for program_id in sorted(
    alias_primary_program_counts
):
    policy = (
        program_policy_registry.get(
            program_id,
            {},
        )
    )

    print(
        "\nPROGRAM:",
        program_id,
    )

    print(
        "Policy:",
        policy.get(
            "policy"
        ),
    )

    print(
        "Primary speaker:",
        policy.get(
            "primary_speaker"
        ),
    )

    print(
        "Default body attribution allowed:",
        policy.get(
            "default_body_attribution_allowed"
        ),
    )

    print(
        "Preserve other turn context:",
        policy.get(
            "preserve_other_turn_context"
        ),
    )

    print(
        "Question-response context required:",
        policy.get(
            "question_response_context_required"
        ),
    )

    print(
        "Unknown speaker inference allowed:",
        policy.get(
            "unknown_speaker_inference_allowed"
        ),
    )

# -------------------------------------------------------------------
# Representative alias-derived examples
# -------------------------------------------------------------------

print(
    "\n=== REPRESENTATIVE ALIAS-DERIVED PRIMARY EXAMPLES ==="
)

for row in alias_primary_rows[:40]:
    print(
        "\n- document_id:",
        row[
            "document_id"
        ],
    )

    print(
        "  program_id:",
        row[
            "program_id"
        ],
    )

    print(
        "  alias:",
        repr(
            row[
                "resolved_primary_alias"
            ]
        ),
    )

    print(
        "  registry canonical:",
        repr(
            primary_aliases_by_program[
                row[
                    "program_id"
                ]
            ][
                row[
                    "resolved_primary_alias"
                ]
            ]
        ),
    )

    print(
        "  text:",
        row[
            "text"
        ]
        .replace(
            "\n",
            " ",
        )[:350],
    )

# -------------------------------------------------------------------
# Validation
# -------------------------------------------------------------------

population_accounted = (
    sum(
        authority_class_counts.values()
    )
    == 16226
)

canonical_population_preserved = (
    authority_class_counts.get(
        "PRIMARY_CANONICAL_ATTRIBUTION",
        0,
    )
    == 11210
)

all_alias_primary_registry_backed = all(
    row[
        "resolved_primary_alias"
    ]
    in primary_aliases_by_program.get(
        row[
            "program_id"
        ],
        {},
    )
    and primary_aliases_by_program[
        row[
            "program_id"
        ]
    ][
        row[
            "resolved_primary_alias"
        ]
    ]
    == PRIMARY_CANONICAL_SPEAKER
    for row in alias_primary_rows
)

no_existing_canonical_overwritten = all(
    (
        row[
            "attribution_speaker"
        ]
        is None
    )
    for row in alias_primary_rows
)

no_unresolved_promoted = all(
    row[
        "effective_label"
    ]
    is not None
    for row
    in alias_primary_rows
)

all_primary_program_policies_consistent = all(
    program_policy_registry.get(
        row[
            "program_id"
        ],
        {},
    ).get(
        "primary_speaker"
    )
    == PRIMARY_CANONICAL_SPEAKER
    for row
    in alias_primary_rows
)

print(
    "\nValidation:"
)

print(
    "Transcript population accounted:",
    "PASS"
    if population_accounted
    else "FAIL",
)

print(
    "Existing canonical primary population preserved:",
    "PASS"
    if canonical_population_preserved
    else "FAIL",
)

print(
    "Every alias-derived primary is registry-backed:",
    "PASS"
    if all_alias_primary_registry_backed
    else "FAIL",
)

print(
    "No existing canonical attribution overwritten:",
    "PASS"
    if no_existing_canonical_overwritten
    else "FAIL",
)

print(
    "No unresolved speaker promoted:",
    "PASS"
    if no_unresolved_promoted
    else "FAIL",
)

print(
    "Alias-derived primary programs agree on primary speaker:",
    "PASS"
    if all_primary_program_policies_consistent
    else "FAIL",
)

cell_gate = (
    population_accounted
    and canonical_population_preserved
    and all_alias_primary_registry_backed
    and no_existing_canonical_overwritten
    and no_unresolved_promoted
    and all_primary_program_policies_consistent
)

print(
    "CODE CELL 27 GATE:",
    "PASS"
    if cell_gate
    else "FAIL",
)
```

<h3>Audit شکاف Alias Registry برای Labelهای صریح دکتر کمیل رودی</h3>

<p>
Code Cell 27 نشان داد که هیچ Segment جدیدی از طریق Program-scoped Alias Registry به Primary Authority ارتقا پیدا نکرد.
اما این نتیجه به معنی نبود Evidence نیست.
</p>

<p>
در Dataset واقعی نمونه‌هایی وجود دارند که
<code>effective.label</code>
آن‌ها دقیقاً
«دکتر کمیل رودی»
است ولی
<code>attribution.speaker</code>
هنوز Null باقی مانده است.
دو Population مهم مشاهده‌شده عبارت‌اند از:
</p>

<ul>
<li><code>program:asre-shirin</code>: تعداد قابل توجهی Label مستقیم «دکتر کمیل رودی».</li>
<li><code>program:khane-to</code>: تعداد قابل توجهی Label مستقیم «دکتر کمیل رودی».</li>
</ul>

<p>
با این حال Alias Registry این Label را برای این دو Program به‌صورت Program-scoped ثبت نکرده است.
بنابراین Cell 27 عمداً آن‌ها را Canonical نکرد.
</p>

<h4>هدف این مرحله</h4>

<p>
باید مشخص شود آیا این وضعیت:
</p>

<ul>
<li>یک Gap در Alias Registry است،</li>
<li>یک تصمیم عمدی Policy است،</li>
<li>یا Evidence دیگری در Program Registry / Human Overlay / Source Metadata وجود دارد که وضعیت را توضیح می‌دهد.</li>
</ul>

<h4>اصل محافظه‌کارانه</h4>

<p>
حتی اگر Label دقیقاً «دکتر کمیل رودی» باشد،
تا زمانی که Contract موجود Repository برای همان Program بررسی نشده است،
هیچ مقدار Canonical در Dataset تغییر نمی‌کند.
</p>

<p>
در این مرحله یک Class توصیفی جدید استفاده می‌شود:
</p>

<p>
<code>PRIMARY_NAME_CANDIDATE_UNREGISTERED_FOR_PROGRAM</code>
</p>

<p>
این Class به معنی Primary Attribution قطعی نیست؛
فقط نشان می‌دهد Source label با نام صریح Primary Subject پروژه سازگار است ولی Alias Registry همان Program هنوز آن را Canonical نکرده است.
</p>

<h4>بررسی‌های این مرحله</h4>

<ul>
<li>نمایش کامل Speaker Policy و Aliasهای هر 13 Program.</li>
<li>کشف Programهایی که Primary Speaker آن‌ها دکتر کمیل رودی است ولی Alias Map ندارند.</li>
<li>شمارش Labelهای صریح مرتبط با دکتر که بدون Program-scoped Alias باقی مانده‌اند.</li>
<li>بررسی <code>base.status</code> و <code>boundary_type</code> این Candidateها.</li>
<li>بررسی Human Attribution Overlayهای موجود.</li>
<li>عدم Promotion یا تغییر Attribution.</li>
</ul>

<h4>معیار PASS</h4>

<p>
تمام Programها باید Audit شوند،
تمام Candidateهای نام صریح باید قابل حسابرسی باشند،
هیچ Candidate نباید به‌طور خودکار Canonical شود
و Population کامل Transcript بدون تغییر حفظ شود.
</p>

</div>

```python
# Code Cell 28

from collections import Counter, defaultdict
import json

# -------------------------------------------------------------------
# Rebuild complete program registry view
# -------------------------------------------------------------------

program_registry_rows = []

for program_entry in program_entries:
    if not isinstance(
        program_entry,
        dict,
    ):
        continue

    program_id = None

    for candidate_key in [
        "program_id",
        "id",
        "program",
    ]:
        candidate_value = (
            program_entry.get(
                candidate_key
            )
        )

        if isinstance(
            candidate_value,
            str,
        ) and candidate_value.strip():
            program_id = candidate_value
            break

    if program_id is None:
        continue

    aliases = program_entry.get(
        "speaker_aliases",
        {}
    )

    if not isinstance(
        aliases,
        dict,
    ):
        aliases = {}

    policy = program_entry.get(
        "speaker_policy",
        {}
    )

    if not isinstance(
        policy,
        dict,
    ):
        policy = {}

    program_registry_rows.append(
        {
            "program_id": program_id,
            "aliases": aliases,
            "policy": policy,
            "raw_entry": program_entry,
        }
    )

program_registry_rows.sort(
    key=lambda row: row[
        "program_id"
    ]
)

print(
    "=== COMPLETE PROGRAM SPEAKER REGISTRY AUDIT ==="
)

print(
    "Programs:",
    len(
        program_registry_rows
    ),
)

for row in program_registry_rows:
    print(
        "\n"
        + "=" * 100
    )

    print(
        "PROGRAM:",
        row["program_id"],
    )

    print(
        "Policy:",
        row[
            "policy"
        ].get(
            "policy"
        ),
    )

    print(
        "Primary speaker:",
        row[
            "policy"
        ].get(
            "primary_speaker"
        ),
    )

    print(
        "Default body attribution allowed:",
        row[
            "policy"
        ].get(
            "default_body_attribution_allowed"
        ),
    )

    print(
        "Preserve other turn context:",
        row[
            "policy"
        ].get(
            "preserve_other_turn_context"
        ),
    )

    print(
        "Question-response context required:",
        row[
            "policy"
        ].get(
            "question_response_context_required"
        ),
    )

    print(
        "Unknown speaker inference allowed:",
        row[
            "policy"
        ].get(
            "unknown_speaker_inference_allowed"
        ),
    )

    print(
        "Speaker aliases:",
        row[
            "aliases"
        ],
    )

# -------------------------------------------------------------------
# Discover explicit source labels that repository already associates
# somewhere with the canonical primary speaker.
# This creates candidate evidence only.
# -------------------------------------------------------------------

known_primary_alias_labels = set()

for aliases in (
    program_alias_registry.values()
):
    for alias, canonical in (
        aliases.items()
    ):
        if (
            canonical
            == PRIMARY_CANONICAL_SPEAKER
        ):
            known_primary_alias_labels.add(
                alias
            )

# Human-confirmed primary names are additional
# repository evidence, but still not cross-program aliases.
human_confirmed_primary_labels = set()

for overlay in human_overlay_records:
    if not isinstance(
        overlay,
        dict,
    ):
        continue

    if (
        overlay.get(
            "primary_dr_roudi_corpus"
        )
        is True
        and overlay.get(
            "human_confirmed_speaker_canonical"
        )
        == PRIMARY_CANONICAL_SPEAKER
    ):
        label = overlay.get(
            "human_confirmed_speaker"
        )

        if isinstance(
            label,
            str,
        ) and label.strip():
            human_confirmed_primary_labels.add(
                label
            )

candidate_primary_name_labels = (
    known_primary_alias_labels
    | human_confirmed_primary_labels
)

print(
    "\n=== REPOSITORY-EVIDENCED PRIMARY NAME LABELS ==="
)

print(
    "Program alias labels:",
    sorted(
        known_primary_alias_labels
    ),
)

print(
    "Human-confirmed primary labels:",
    sorted(
        human_confirmed_primary_labels
    ),
)

print(
    "Combined candidate labels:",
    sorted(
        candidate_primary_name_labels
    ),
)

# -------------------------------------------------------------------
# Identify labels matching known primary-name evidence
# but lacking program-scoped canonical alias.
# -------------------------------------------------------------------

unregistered_primary_name_candidates = []

for row in speaker_semantic_rows:
    if (
        row[
            "attribution_speaker"
        ]
        is not None
    ):
        continue

    effective_label = row[
        "effective_label"
    ]

    if (
        not isinstance(
            effective_label,
            str,
        )
        or effective_label
        not in candidate_primary_name_labels
    ):
        continue

    program_id = row[
        "program_id"
    ]

    scoped_aliases = (
        program_alias_registry.get(
            program_id,
            {},
        )
    )

    scoped_canonical = (
        scoped_aliases.get(
            effective_label
        )
    )

    if (
        scoped_canonical
        == PRIMARY_CANONICAL_SPEAKER
    ):
        # Already handled by Cell 27.
        continue

    unregistered_primary_name_candidates.append(
        {
            **row,
            "candidate_class": (
                "PRIMARY_NAME_CANDIDATE_UNREGISTERED_FOR_PROGRAM"
            ),
            "scoped_alias_value": (
                scoped_canonical
            ),
        }
    )

print(
    "\n=== UNREGISTERED PRIMARY-NAME CANDIDATES ==="
)

print(
    "Candidate segments:",
    len(
        unregistered_primary_name_candidates
    ),
)

candidate_program_counts = Counter(
    row[
        "program_id"
    ]
    for row in (
        unregistered_primary_name_candidates
    )
)

candidate_label_counts = Counter(
    (
        row[
            "program_id"
        ],
        row[
            "effective_label"
        ],
    )
    for row in (
        unregistered_primary_name_candidates
    )
)

print(
    "\nBy program:"
)

for program_id, count in (
    candidate_program_counts.most_common()
):
    print(
        "-",
        program_id,
        ":",
        count,
    )

print(
    "\nBy program / label:"
)

for (
    program_id,
    label,
), count in (
    candidate_label_counts.most_common()
):
    print(
        "-",
        program_id,
        "|",
        repr(label),
        ":",
        count,
    )

# -------------------------------------------------------------------
# Evidence quality inside candidates
# -------------------------------------------------------------------

candidate_base_status_counts = Counter(
    row[
        "base_status"
    ]
    for row in (
        unregistered_primary_name_candidates
    )
)

candidate_boundary_counts = Counter(
    row[
        "boundary_type"
    ]
    for row in (
        unregistered_primary_name_candidates
    )
)

print(
    "\n=== CANDIDATE SOURCE EVIDENCE PROFILE ==="
)

print(
    "Base status:"
)

for value, count in (
    candidate_base_status_counts.most_common()
):
    print(
        "-",
        repr(value),
        ":",
        count,
    )

print(
    "\nBoundary type:"
)

for value, count in (
    candidate_boundary_counts.most_common()
):
    print(
        "-",
        repr(value),
        ":",
        count,
    )

# -------------------------------------------------------------------
# Program policies for candidate populations
# -------------------------------------------------------------------

print(
    "\n=== CANDIDATE PROGRAM POLICY CONTEXT ==="
)

program_row_lookup = {
    row[
        "program_id"
    ]: row
    for row in program_registry_rows
}

for program_id in sorted(
    candidate_program_counts
):
    registry_row = (
        program_row_lookup.get(
            program_id
        )
    )

    print(
        "\nPROGRAM:",
        program_id,
    )

    print(
        "Candidate count:",
        candidate_program_counts[
            program_id
        ],
    )

    if registry_row is None:
        print(
            "Registry row: MISSING"
        )
        continue

    policy = registry_row[
        "policy"
    ]

    print(
        "Policy:",
        policy.get(
            "policy"
        ),
    )

    print(
        "Primary speaker:",
        policy.get(
            "primary_speaker"
        ),
    )

    print(
        "Default body attribution allowed:",
        policy.get(
            "default_body_attribution_allowed"
        ),
    )

    print(
        "Preserve other turn context:",
        policy.get(
            "preserve_other_turn_context"
        ),
    )

    print(
        "Question-response context required:",
        policy.get(
            "question_response_context_required"
        ),
    )

    print(
        "Unknown speaker inference allowed:",
        policy.get(
            "unknown_speaker_inference_allowed"
        ),
    )

    print(
        "Current aliases:",
        registry_row[
            "aliases"
        ],
    )

# -------------------------------------------------------------------
# Human overlay detail
# -------------------------------------------------------------------

print(
    "\n=== HUMAN ATTRIBUTION OVERLAY DETAILS ==="
)

for index, overlay in enumerate(
    human_overlay_records,
    start=1,
):
    print(
        "\nOVERLAY:",
        index,
    )

    print(
        json.dumps(
            overlay,
            ensure_ascii=False,
            indent=2,
            sort_keys=True,
        )
    )

# -------------------------------------------------------------------
# Representative candidate examples
# -------------------------------------------------------------------

print(
    "\n=== REPRESENTATIVE UNREGISTERED PRIMARY-NAME CANDIDATES ==="
)

examples_by_candidate_program = defaultdict(
    list
)

for row in (
    unregistered_primary_name_candidates
):
    program_id = row[
        "program_id"
    ]

    if len(
        examples_by_candidate_program[
            program_id
        ]
    ) < 15:
        examples_by_candidate_program[
            program_id
        ].append(
            row
        )

for program_id in sorted(
    examples_by_candidate_program
):
    print(
        "\n"
        + "=" * 100
    )

    print(
        "PROGRAM:",
        program_id,
    )

    for row in (
        examples_by_candidate_program[
            program_id
        ]
    ):
        print(
            "\n- document_id:",
            row[
                "document_id"
            ],
        )

        print(
            "  label:",
            repr(
                row[
                    "effective_label"
                ]
            ),
        )

        print(
            "  attribution:",
            repr(
                row[
                    "attribution_speaker"
                ]
            ),
        )

        print(
            "  base status:",
            repr(
                row[
                    "base_status"
                ]
            ),
        )

        print(
            "  boundary type:",
            repr(
                row[
                    "boundary_type"
                ]
            ),
        )

        print(
            "  scoped alias value:",
            repr(
                row[
                    "scoped_alias_value"
                ]
            ),
        )

        print(
            "  text:",
            row[
                "text"
            ]
            .replace(
                "\n",
                " ",
            )[:400],
        )

# -------------------------------------------------------------------
# Programs where primary speaker is known but no aliases are registered
# -------------------------------------------------------------------

print(
    "\n=== PRIMARY-SPEAKER PROGRAMS WITHOUT ALIAS MAP ==="
)

primary_programs_without_aliases = []

for row in program_registry_rows:
    policy = row[
        "policy"
    ]

    if (
        policy.get(
            "primary_speaker"
        )
        == PRIMARY_CANONICAL_SPEAKER
        and len(
            row[
                "aliases"
            ]
        )
        == 0
    ):
        primary_programs_without_aliases.append(
            row[
                "program_id"
            ]
        )

for program_id in (
    primary_programs_without_aliases
):
    print(
        "-",
        program_id,
    )

print(
    "Count:",
    len(
        primary_programs_without_aliases
    ),
)

# -------------------------------------------------------------------
# No-modification validation
# -------------------------------------------------------------------

all_programs_accounted = (
    len(
        program_registry_rows
    )
    == speaker_alias_data.get(
        "program_count"
    )
)

candidate_population_finite = (
    len(
        unregistered_primary_name_candidates
    )
    >= 0
)

all_candidates_currently_unattributed = all(
    row[
        "attribution_speaker"
    ]
    is None
    for row in (
        unregistered_primary_name_candidates
    )
)

all_candidates_not_scoped_primary_alias = all(
    row[
        "scoped_alias_value"
    ]
    != PRIMARY_CANONICAL_SPEAKER
    for row in (
        unregistered_primary_name_candidates
    )
)

all_candidates_have_source_label = all(
    isinstance(
        row[
            "effective_label"
        ],
        str,
    )
    and row[
        "effective_label"
    ].strip()
    for row in (
        unregistered_primary_name_candidates
    )
)

transcript_population_preserved = (
    len(
        speaker_semantic_rows
    )
    == 16226
)

print(
    "\nValidation:"
)

print(
    "All registry programs accounted:",
    "PASS"
    if all_programs_accounted
    else "FAIL",
)

print(
    "Candidate audit executed:",
    "PASS"
    if candidate_population_finite
    else "FAIL",
)

print(
    "All candidates currently lack canonical attribution:",
    "PASS"
    if all_candidates_currently_unattributed
    else "FAIL",
)

print(
    "No candidate already has program-scoped primary alias:",
    "PASS"
    if all_candidates_not_scoped_primary_alias
    else "FAIL",
)

print(
    "All candidates have explicit source label:",
    "PASS"
    if all_candidates_have_source_label
    else "FAIL",
)

print(
    "Transcript population preserved:",
    "PASS"
    if transcript_population_preserved
    else "FAIL",
)

cell_gate = (
    all_programs_accounted
    and candidate_population_finite
    and all_candidates_currently_unattributed
    and all_candidates_not_scoped_primary_alias
    and all_candidates_have_source_label
    and transcript_population_preserved
)

print(
    "CODE CELL 28 GATE:",
    "PASS"
    if cell_gate
    else "FAIL",
)
```

<h4>Human Attribution Overlay</h4>

<p>
Repository همچنین دو Human Attribution Overlay ثبت کرده است:
</p>

<ul>
<li>یک مورد برای Secondary Guest.</li>
<li>یک مورد برای Primary Subject یعنی دکتر کمیل رودی.</li>

<p>
قبل از تعریف Authority Contract نهایی باید بررسی شود که این Overlayها در Dataset فعلی چگونه بازتاب یافته‌اند:

آیا هنوز Segment مربوطه Unresolved است،
یا Evidence فقط در Registry باقی مانده است.
</p>

<h4>Evidence Classهای این مرحله</h4>

<ul>
<li><code>PRIMARY_CANONICAL_ATTRIBUTION</code>: Canonical Attribution موجود به دکتر.</li>
<li><code>PRIMARY_EXPLICIT_SOURCE_MARKER</code>: Label صریح Source دقیقاً دکتر کمیل رودی است و Program نیز ایشان را Primary Speaker معرفی کرده است.</li>
<li><code>PRIMARY_HUMAN_CONFIRMED_OVERLAY</code>: Human Review صریحاً دکتر را برای همان Utterance تأیید کرده است.</li>
<li><code>SECONDARY_HUMAN_CONFIRMED_OVERLAY</code>: Human Review Speaker دیگری را به‌عنوان Guest/Secondary تأیید کرده است.</li>
<li><code>NONPRIMARY_EXPLICIT_CONTEXT</code>: مجری، مهمان، Participant یا Speaker صریح دیگر.</li>
<li><code>UNRESOLVED_SPEAKER_CONTEXT</code>: Speaker بدون Evidence کافی باقی مانده است.</li>
</ul>

<h4>اصل مهم</h4>

<p>
این Classها برای Authority و Retrieval Evidence هستند و نباید Source Attribution را تحریف کنند.
برای Programهای چندنفره، Context سؤال و پاسخ همچنان باید حفظ شود.
</p>

<p>
همچنین حتی Human-confirmed Primary Evidence در این مرحله باعث Mutation فایل ورودی نمی‌شود.
</p>

<h4>معیار PASS</h4>

<p>
هر Human Overlay باید دقیقاً به Document موجود متصل شود،
718 Source Marker صریح باید مستقل بازتولید شوند،
هیچ Guest به Primary تبدیل نشود،
و تمام 16,226 Transcript بدون حذف یا تغییر در یک Authority Evidence Class قرار گیرند.
</p>

```python
# Code Cell 29

from collections import Counter, defaultdict
import json

# -------------------------------------------------------------------
# Human overlay lookup
# -------------------------------------------------------------------

overlay_lookup = {}

for overlay in human_overlay_records:
    if not isinstance(
        overlay,
        dict,
    ):
        continue

    utterance_id = overlay.get(
        "utterance_id"
    )

    if not isinstance(
        utterance_id,
        str,
    ):
        continue

    overlay_lookup[
        utterance_id
    ] = overlay

print(
    "=== HUMAN ATTRIBUTION OVERLAY DATASET CHECK ==="
)

print(
    "Overlay records:",
    len(
        overlay_lookup
    ),
)

dataset_document_lookup = {
    row[
        "document_id"
    ]: row
    for row in speaker_semantic_rows
}

overlay_dataset_rows = []

for utterance_id, overlay in (
    overlay_lookup.items()
):
    dataset_row = (
        dataset_document_lookup.get(
            utterance_id
        )
    )

    overlay_dataset_rows.append(
        {
            "utterance_id": (
                utterance_id
            ),
            "overlay": overlay,
            "dataset_row": (
                dataset_row
            ),
        }
    )

for item in overlay_dataset_rows:
    overlay = item[
        "overlay"
    ]

    row = item[
        "dataset_row"
    ]

    print(
        "\n"
        + "=" * 100
    )

    print(
        "Utterance:",
        item[
            "utterance_id"
        ],
    )

    print(
        "Dataset match:",
        row is not None,
    )

    print(
        "Human confirmed speaker:",
        repr(
            overlay.get(
                "human_confirmed_speaker"
            )
        ),
    )

    print(
        "Human canonical:",
        repr(
            overlay.get(
                "human_confirmed_speaker_canonical"
            )
        ),
    )

    print(
        "Primary Dr. Roudi corpus:",
        overlay.get(
            "primary_dr_roudi_corpus"
        ),
    )

    print(
        "Role:",
        overlay.get(
            "speaker_role_for_project"
        ),
    )

    print(
        "Basis:",
        overlay.get(
            "human_confirmation_basis"
        ),
    )

    if row is not None:
        print(
            "Current attribution:",
            repr(
                row[
                    "attribution_speaker"
                ]
            ),
        )

        print(
            "Effective canonical:",
            repr(
                row[
                    "effective_canonical"
                ]
            ),
        )

        print(
            "Effective label:",
            repr(
                row[
                    "effective_label"
                ]
            ),
        )

        print(
            "Base status:",
            repr(
                row[
                    "base_status"
                ]
            ),
        )

        print(
            "Boundary type:",
            repr(
                row[
                    "boundary_type"
                ]
            ),
        )

        print(
            "Current semantic class:",
            row[
                "evidence_class"
            ],
        )

        print(
            "Text:",
            row[
                "text"
            ]
            .replace(
                "\n",
                " ",
            )[:500],
        )

# -------------------------------------------------------------------
# Explicit primary source-marker evidence
# -------------------------------------------------------------------

print(
    "\n=== EXPLICIT PRIMARY SOURCE-MARKER DISCOVERY ==="
)

explicit_primary_source_rows = []

for row in speaker_semantic_rows:
    if (
        row[
            "attribution_speaker"
        ]
        is not None
    ):
        continue

    program_id = row[
        "program_id"
    ]

    registry_row = (
        program_row_lookup.get(
            program_id
        )
    )

    if registry_row is None:
        continue

    primary_speaker = (
        registry_row[
            "policy"
        ].get(
            "primary_speaker"
        )
    )

    if (
        primary_speaker
        != PRIMARY_CANONICAL_SPEAKER
    ):
        continue

    if (
        row[
            "effective_label"
        ]
        != "دکتر کمیل رودی"
    ):
        continue

    if (
        row[
            "base_status"
        ]
        != "EXPLICIT_SOURCE_MARKER"
    ):
        continue

    explicit_primary_source_rows.append(
        row
    )

print(
    "Explicit primary source-marker segments:",
    len(
        explicit_primary_source_rows
    ),
)

explicit_primary_program_counts = Counter(
    row[
        "program_id"
    ]
    for row in explicit_primary_source_rows
)

for program_id, count in (
    explicit_primary_program_counts.most_common()
):
    print(
        "-",
        program_id,
        ":",
        count,
    )

explicit_primary_boundary_counts = Counter(
    row[
        "boundary_type"
    ]
    for row in explicit_primary_source_rows
)

print(
    "Boundary types:",
    dict(
        explicit_primary_boundary_counts
    ),
)

# -------------------------------------------------------------------
# Overlay-aware authority classification
# No mutation of input records.
# -------------------------------------------------------------------

def classify_final_transcript_authority(
    row,
):
    document_id = row[
        "document_id"
    ]

    overlay = overlay_lookup.get(
        document_id
    )

    # Human-reviewed evidence takes precedence
    # for the specific utterance.
    if overlay is not None:
        human_canonical = (
            overlay.get(
                "human_confirmed_speaker_canonical"
            )
        )

        primary_flag = (
            overlay.get(
                "primary_dr_roudi_corpus"
            )
        )

        if (
            human_canonical
            == PRIMARY_CANONICAL_SPEAKER
            and primary_flag is True
        ):
            return {
                "authority_class": (
                    "PRIMARY_HUMAN_CONFIRMED_OVERLAY"
                ),
                "authority_basis": (
                    "DIRECT_HUMAN_REVIEW"
                ),
            }

        return {
            "authority_class": (
                "SECONDARY_HUMAN_CONFIRMED_OVERLAY"
            ),
            "authority_basis": (
                "DIRECT_HUMAN_REVIEW"
            ),
        }

    # Existing canonical document attribution.
    if (
        row[
            "attribution_speaker"
        ]
        == PRIMARY_CANONICAL_SPEAKER
    ):
        return {
            "authority_class": (
                "PRIMARY_CANONICAL_ATTRIBUTION"
            ),
            "authority_basis": (
                "CURRENT_DOCUMENT_ATTRIBUTION"
            ),
        }

    # Preserve any explicit non-primary
    # canonical attribution if it ever appears.
    if (
        row[
            "attribution_speaker"
        ]
        is not None
    ):
        return {
            "authority_class": (
                "NONPRIMARY_EXPLICIT_CONTEXT"
            ),
            "authority_basis": (
                "CURRENT_DOCUMENT_NONPRIMARY_ATTRIBUTION"
            ),
        }

    program_id = row[
        "program_id"
    ]

    registry_row = (
        program_row_lookup.get(
            program_id
        )
    )

    program_primary = None

    if registry_row is not None:
        program_primary = (
            registry_row[
                "policy"
            ].get(
                "primary_speaker"
            )
        )

    # Direct source marker naming Dr. Komeil Roudi
    # is evidence, not alias inference.
    if (
        program_primary
        == PRIMARY_CANONICAL_SPEAKER
        and row[
            "effective_label"
        ]
        == "دکتر کمیل رودی"
        and row[
            "base_status"
        ]
        == "EXPLICIT_SOURCE_MARKER"
    ):
        return {
            "authority_class": (
                "PRIMARY_EXPLICIT_SOURCE_MARKER"
            ),
            "authority_basis": (
                "EXPLICIT_SOURCE_SPEAKER_LABEL"
            ),
        }

    # Other explicit labels remain independent context.
    if (
        isinstance(
            row[
                "effective_label"
            ],
            str,
        )
        and row[
            "effective_label"
        ].strip()
    ):
        return {
            "authority_class": (
                "NONPRIMARY_EXPLICIT_CONTEXT"
            ),
            "authority_basis": (
                "SOURCE_OBSERVED_NONPRIMARY_OR_UNCANONICAL_LABEL"
            ),
        }

    return {
        "authority_class": (
            "UNRESOLVED_SPEAKER_CONTEXT"
        ),
        "authority_basis": (
            "NO_SUFFICIENT_SPEAKER_EVIDENCE"
        ),
    }

# -------------------------------------------------------------------
# Classify all Transcript segments
# -------------------------------------------------------------------

final_transcript_authority_rows = []

for row in speaker_semantic_rows:
    classification = (
        classify_final_transcript_authority(
            row
        )
    )

    final_transcript_authority_rows.append(
        {
            **row,
            **classification,
        }
    )

final_authority_class_counts = Counter(
    row[
        "authority_class"
    ]
    for row
    in final_transcript_authority_rows
)

final_authority_basis_counts = Counter(
    row[
        "authority_basis"
    ]
    for row
    in final_transcript_authority_rows
)

print(
    "\n=== FINAL DESCRIPTIVE TRANSCRIPT AUTHORITY PROFILE ==="
)

authority_class_order = [
    "PRIMARY_CANONICAL_ATTRIBUTION",
    "PRIMARY_EXPLICIT_SOURCE_MARKER",
    "PRIMARY_HUMAN_CONFIRMED_OVERLAY",
    "SECONDARY_HUMAN_CONFIRMED_OVERLAY",
    "NONPRIMARY_EXPLICIT_CONTEXT",
    "UNRESOLVED_SPEAKER_CONTEXT",
]

for authority_class in (
    authority_class_order
):
    print(
        "-",
        authority_class,
        ":",
        final_authority_class_counts.get(
            authority_class,
            0,
        ),
    )

print(
    "\nAuthority basis:"
)

for basis, count in (
    final_authority_basis_counts.most_common()
):
    print(
        "-",
        basis,
        ":",
        count,
    )

# -------------------------------------------------------------------
# Primary / contextual / unresolved aggregate
# -------------------------------------------------------------------

primary_authority_classes = {
    "PRIMARY_CANONICAL_ATTRIBUTION",
    "PRIMARY_EXPLICIT_SOURCE_MARKER",
    "PRIMARY_HUMAN_CONFIRMED_OVERLAY",
}

primary_evidence_count = sum(
    final_authority_class_counts.get(
        authority_class,
        0,
    )
    for authority_class in (
        primary_authority_classes
    )
)

contextual_nonprimary_count = sum(
    final_authority_class_counts.get(
        authority_class,
        0,
    )
    for authority_class in [
        "SECONDARY_HUMAN_CONFIRMED_OVERLAY",
        "NONPRIMARY_EXPLICIT_CONTEXT",
    ]
)

unresolved_count = (
    final_authority_class_counts.get(
        "UNRESOLVED_SPEAKER_CONTEXT",
        0,
    )
)

print(
    "\n=== AGGREGATED TRANSCRIPT AUTHORITY EVIDENCE ==="
)

print(
    "Primary Dr. Roudi evidence:",
    primary_evidence_count,
)

print(
    "Explicit secondary/context evidence:",
    contextual_nonprimary_count,
)

print(
    "Unresolved speaker context:",
    unresolved_count,
)

print(
    "Total:",
    (
        primary_evidence_count
        + contextual_nonprimary_count
        + unresolved_count
    ),
)

# -------------------------------------------------------------------
# Program-level final profile
# -------------------------------------------------------------------

print(
    "\n=== PROGRAM-LEVEL FINAL AUTHORITY PROFILE ==="
)

program_final_authority = defaultdict(
    Counter
)

for row in (
    final_transcript_authority_rows
):
    program_final_authority[
        row[
            "program_id"
        ]
    ][
        row[
            "authority_class"
        ]
    ] += 1

for program_id in sorted(
    program_final_authority
):
    counts = (
        program_final_authority[
            program_id
        ]
    )

    print(
        "\nPROGRAM:",
        program_id,
    )

    print(
        "Total:",
        sum(
            counts.values()
        ),
    )

    for authority_class in (
        authority_class_order
    ):
        count = counts.get(
            authority_class,
            0,
        )

        if count:
            print(
                "-",
                authority_class,
                ":",
                count,
            )

# -------------------------------------------------------------------
# Validation
# -------------------------------------------------------------------

all_overlays_match_dataset = all(
    item[
        "dataset_row"
    ]
    is not None
    for item in overlay_dataset_rows
)

overlay_count_reproduced = (
    len(
        overlay_dataset_rows
    )
    == len(
        human_overlay_records
    )
)

explicit_source_population_reproduced = (
    len(
        explicit_primary_source_rows
    )
    == 718
)

explicit_source_programs_reproduced = (
    explicit_primary_program_counts
    == Counter(
        {
            "program:asre-shirin": 440,
            "program:khane-to": 278,
        }
    )
)

population_accounted = (
    sum(
        final_authority_class_counts.values()
    )
    == 16226
)

no_secondary_overlay_promoted = all(
    not (
        row[
            "authority_class"
        ]
        in primary_authority_classes
        and row[
            "document_id"
        ]
        in overlay_lookup
        and overlay_lookup[
            row[
                "document_id"
            ]
        ].get(
            "primary_dr_roudi_corpus"
        )
        is False
    )
    for row in (
        final_transcript_authority_rows
    )
)

no_input_mutation = (
    len(
        speaker_semantic_rows
    )
    == 16226
    and len(
        quality_records
    )
    == 19456
)

print(
    "\nValidation:"
)

print(
    "All Human Overlays match dataset documents:",
    "PASS"
    if all_overlays_match_dataset
    else "FAIL",
)

print(
    "Human Overlay population reproduced:",
    "PASS"
    if overlay_count_reproduced
    else "FAIL",
)

print(
    "718 explicit primary source markers reproduced:",
    "PASS"
    if explicit_source_population_reproduced
    else "FAIL",
)

print(
    "Explicit source-marker program counts reproduced:",
    "PASS"
    if explicit_source_programs_reproduced
    else "FAIL",
)

print(
    "Full Transcript population accounted:",
    "PASS"
    if population_accounted
    else "FAIL",
)

print(
    "Secondary Human Overlay never promoted to Primary:",
    "PASS"
    if no_secondary_overlay_promoted
    else "FAIL",
)

print(
    "Input populations preserved without mutation:",
    "PASS"
    if no_input_mutation
    else "FAIL",
)

cell_gate = (
    all_overlays_match_dataset
    and overlay_count_reproduced
    and explicit_source_population_reproduced
    and explicit_source_programs_reproduced
    and population_accounted
    and no_secondary_overlay_promoted
    and no_input_mutation
)

print(
    "CODE CELL 29 GATE:",
    "PASS"
    if cell_gate
    else "FAIL",
)
```

<h4>تفاوت این Validation با بررسی‌های قبلی</h4>

<p>
بررسی‌های قبلی عمدتاً Field Coverage و Nullability را تحلیل می‌کردند.
در این مرحله خود JSON Schema اجرا می‌شود؛ بنابراین مواردی مانند:

اما Metadata gapهای Schema-valid مانند
<code>language = null</code>
یا Rights با وضعیت
<code>EVIDENCE_REQUIRED</code>
به‌تنهایی Schema failure محسوب نمی‌شوند.
</p>

<p>
همچنین در این مرحله هنوز هیچ Document حذف یا Quarantine نمی‌شود.
فقط Population واقعی Schema-valid و Schema-invalid اندازه‌گیری می‌شود.
</p>

<h4>خروجی مورد انتظار</h4>

<ul>
<li>Validator واقعی مورد استفاده.</li>
<li>تعداد رکوردهای Valid و Invalid.</li>
<li>توزیع Invalidها به تفکیک Source.</li>
<li>توزیع Error Validatorها مانند <code>required</code>، <code>type</code>، <code>enum</code> و غیره.</li>
<li>مسیر دقیق Schema Error در صورت وجود.</li>
<li>نمونه محدود از رکوردهای Invalid برای بررسی انسانی.</li>
</ul>

<h4>معیار PASS</h4>

<p>
هر 19,456 Document باید دقیقاً یک‌بار Validate شود،
هیچ Validation exception پردازش‌نشده‌ای رخ ندهد
و Population نهایی Valid + Invalid دقیقاً برابر Dataset ورودی باشد.
</p>

<p>
PASS این Cell الزاماً به معنی صفر Schema Error نیست؛
بلکه یعنی Audit کامل و قابل اتکا انجام شده است.
اگر Schema Error وجود داشته باشد، قبل از Policy نهایی باید همان Population جداگانه بررسی شود.

```python
# Code Cell 30

from collections import Counter, defaultdict
import importlib.util
import json

# -------------------------------------------------------------------
# JSON Schema validator discovery
# -------------------------------------------------------------------

jsonschema_available = (
    importlib.util.find_spec(
        "jsonschema"
    )
    is not None
)

print(
    "=== JSON SCHEMA VALIDATOR DISCOVERY ==="
)

print(
    "jsonschema available:",
    jsonschema_available,
)

schema_validation_exception = None

if jsonschema_available:
    import jsonschema

    from jsonschema.validators import (
        validator_for,
    )

    print(
        "jsonschema version:",
        getattr(
            jsonschema,
            "__version__",
            "UNKNOWN",
        ),
    )

    ValidatorClass = validator_for(
        document_schema
    )

    print(
        "Selected validator:",
        ValidatorClass.__name__,
    )

    # Validate the schema itself first.
    try:
        ValidatorClass.check_schema(
            document_schema
        )

        schema_definition_valid = True

    except Exception as exc:
        schema_definition_valid = False
        schema_validation_exception = exc

else:
    ValidatorClass = None
    schema_definition_valid = False

print(
    "Schema definition valid:",
    (
        "PASS"
        if schema_definition_valid
        else "FAIL"
    ),
)

if schema_validation_exception is not None:
    print(
        "Schema validation exception:",
        repr(
            schema_validation_exception
        ),
    )

# -------------------------------------------------------------------
# Full-document schema validation
# -------------------------------------------------------------------

schema_valid_document_count = 0
schema_invalid_document_count = 0

schema_invalid_by_source = Counter()
schema_error_validator_counts = Counter()
schema_error_path_counts = Counter()

schema_invalid_records = []

validation_runtime_exception = None

if (
    jsonschema_available
    and schema_definition_valid
):
    validator = ValidatorClass(
        document_schema
    )

    try:
        for record_index, record in enumerate(
            quality_records
        ):
            errors = sorted(
                validator.iter_errors(
                    record
                ),
                key=lambda error: (
                    list(
                        error.absolute_path
                    ),
                    error.validator
                    or "",
                    error.message,
                ),
            )

            if not errors:
                schema_valid_document_count += 1
                continue

            schema_invalid_document_count += 1

            source_type = record.get(
                "source_type",
                "UNKNOWN",
            )

            schema_invalid_by_source[
                source_type
            ] += 1

            error_summaries = []

            for error in errors:
                validator_name = (
                    error.validator
                    or "UNKNOWN"
                )

                schema_error_validator_counts[
                    validator_name
                ] += 1

                absolute_path_parts = [
                    str(part)
                    for part
                    in error.absolute_path
                ]

                if absolute_path_parts:
                    absolute_path = (
                        "$."
                        + ".".join(
                            absolute_path_parts
                        )
                    )
                else:
                    absolute_path = "$"

                schema_error_path_counts[
                    absolute_path
                ] += 1

                error_summaries.append(
                    {
                        "path": absolute_path,
                        "validator": (
                            validator_name
                        ),
                        "message": (
                            error.message
                        ),
                        "validator_value": (
                            error.validator_value
                        ),
                    }
                )

            if len(
                schema_invalid_records
            ) < 50:
                schema_invalid_records.append(
                    {
                        "record_index": (
                            record_index
                        ),
                        "document_id": (
                            record.get(
                                "document_id"
                            )
                        ),
                        "source_type": (
                            source_type
                        ),
                        "source_id": (
                            record.get(
                                "source_id"
                            )
                        ),
                        "error_count": len(
                            errors
                        ),
                        "errors": (
                            error_summaries
                        ),
                    }
                )

    except Exception as exc:
        validation_runtime_exception = exc

# -------------------------------------------------------------------
# Report
# -------------------------------------------------------------------

print(
    "\n=== FULL DATASET JSON SCHEMA VALIDATION ==="
)

print(
    "Documents expected:",
    len(
        quality_records
    ),
)

print(
    "Schema-valid documents:",
    schema_valid_document_count,
)

print(
    "Schema-invalid documents:",
    schema_invalid_document_count,
)

print(
    "Validation runtime exception:",
    (
        "NONE"
        if validation_runtime_exception
        is None
        else repr(
            validation_runtime_exception
        )
    ),
)

print(
    "\n=== INVALID DOCUMENTS BY SOURCE ==="
)

for source_type in [
    "website",
    "book",
    "transcript",
    "UNKNOWN",
]:
    count = (
        schema_invalid_by_source.get(
            source_type,
            0,
        )
    )

    if count:
        print(
            "-",
            source_type,
            ":",
            count,
        )

if not schema_invalid_by_source:
    print(
        "- NONE"
    )

print(
    "\n=== SCHEMA ERROR VALIDATOR DISTRIBUTION ==="
)

if schema_error_validator_counts:
    for validator_name, count in (
        schema_error_validator_counts.most_common()
    ):
        print(
            "-",
            validator_name,
            ":",
            count,
        )
else:
    print(
        "- NONE"
    )

print(
    "\n=== MOST COMMON SCHEMA ERROR PATHS ==="
)

if schema_error_path_counts:
    for path, count in (
        schema_error_path_counts.most_common(
            40
        )
    ):
        print(
            "-",
            path,
            ":",
            count,
        )
else:
    print(
        "- NONE"
    )

# -------------------------------------------------------------------
# Limited invalid examples
# -------------------------------------------------------------------

print(
    "\n=== INVALID DOCUMENT EXAMPLES ==="
)

if not schema_invalid_records:
    print(
        "- NONE"
    )

for item in schema_invalid_records:
    print(
        "\n"
        + "=" * 100
    )

    print(
        "Record index:",
        item[
            "record_index"
        ],
    )

    print(
        "Document ID:",
        item[
            "document_id"
        ],
    )

    print(
        "Source type:",
        item[
            "source_type"
        ],
    )

    print(
        "Source ID:",
        item[
            "source_id"
        ],
    )

    print(
        "Schema errors:",
        item[
            "error_count"
        ],
    )

    for error in item[
        "errors"
    ]:
        print(
            " - path:",
            error[
                "path"
            ],
        )

        print(
            "   validator:",
            error[
                "validator"
            ],
        )

        print(
            "   message:",
            error[
                "message"
            ],
        )

        print(
            "   validator value:",
            repr(
                error[
                    "validator_value"
                ]
            ),
        )

# -------------------------------------------------------------------
# Independent population consistency
# -------------------------------------------------------------------

validated_population = (
    schema_valid_document_count
    + schema_invalid_document_count
)

population_accounted = (
    validated_population
    == len(
        quality_records
    )
)

source_population_accounted = (
    sum(
        schema_invalid_by_source.values()
    )
    == schema_invalid_document_count
)

runtime_clean = (
    validation_runtime_exception
    is None
)

print(
    "\n=== STRUCTURAL QUALITY INTERPRETATION ==="
)

if (
    schema_invalid_document_count
    == 0
    and population_accounted
):
    print(
        "All documents are schema-valid."
    )

    print(
        "No hard structural schema failure "
        "was discovered in the normalized dataset."
    )

elif population_accounted:
    print(
        "Schema-invalid documents exist."
    )

    print(
        "These records require separate "
        "hard-failure inspection before "
        "Quality/Quarantine policy is finalized."
    )

else:
    print(
        "Schema validation population is incomplete."
    )

# -------------------------------------------------------------------
# Validation gate
# -------------------------------------------------------------------

print(
    "\nValidation:"
)

print(
    "jsonschema validator available:",
    "PASS"
    if jsonschema_available
    else "FAIL",
)

print(
    "Schema definition accepted by validator:",
    "PASS"
    if schema_definition_valid
    else "FAIL",
)

print(
    "Validation completed without runtime exception:",
    "PASS"
    if runtime_clean
    else "FAIL",
)

print(
    "All dataset documents accounted:",
    "PASS"
    if population_accounted
    else "FAIL",
)

print(
    "Invalid source population accounted:",
    "PASS"
    if source_population_accounted
    else "FAIL",
)

cell_gate = (
    jsonschema_available
    and schema_definition_valid
    and runtime_clean
    and population_accounted
    and source_population_accounted
)

print(
    "CODE CELL 30 GATE:",
    "PASS"
    if cell_gate
    else "FAIL",
)
```

<h3>Audit ساختار انسانی تأییدشده Programهای «خانه تو» و «عصر شیرین»</h3>

```python
# Code Cell 31

from collections import Counter, defaultdict
import json

# -------------------------------------------------------------------
# Human-confirmed project context
# -------------------------------------------------------------------

human_confirmed_program_context = {
    "program:khane-to": {
        "evidence_type": (
            "HUMAN_CONFIRMED_PROJECT_CONTEXT"
        ),
        "structure": (
            "TWO_PERSON_HOST_AND_DR_ROUDI"
        ),
        "primary_speaker": (
            "Dr. Komeil Roudi"
        ),
        "host_count": 1,
        "participant_group_expected": False,
    },
    "program:asre-shirin": {
        "evidence_type": (
            "HUMAN_CONFIRMED_PROJECT_CONTEXT"
        ),
        "structure": (
            "HOST_DR_ROUDI_AND_PARTICIPANTS"
        ),
        "primary_speaker": (
            "Dr. Komeil Roudi"
        ),
        "season_count": 2,
        "host_count": 1,
        "approx_participant_count": 8,
        "host_gender": "female",
        "participant_gender": "female",
    },
}

print(
    "=== HUMAN-CONFIRMED PROGRAM CONTEXT ==="
)

print(
    json.dumps(
        human_confirmed_program_context,
        ensure_ascii=False,
        indent=2,
        sort_keys=True,
    )
)

# -------------------------------------------------------------------
# Target program records
# -------------------------------------------------------------------

target_program_ids = [
    "program:khane-to",
    "program:asre-shirin",
]

target_program_records = {
    program_id: [
        record
        for record in transcript_records_current
        if (
            record.get(
                "source_metadata",
                {}
            ).get(
                "program_id"
            )
            == program_id
        )
    ]
    for program_id
    in target_program_ids
}

print(
    "\n=== TARGET PROGRAM POPULATIONS ==="
)

for program_id in target_program_ids:
    records = (
        target_program_records[
            program_id
        ]
    )

    print(
        "-",
        program_id,
        ":",
        len(records),
    )

# -------------------------------------------------------------------
# Speaker structure extraction
# -------------------------------------------------------------------

def get_effective_speaker_fields(
    record,
):
    speaker_object = (
        record.get(
            "source_metadata",
            {}
        ).get(
            "speaker",
            {}
        )
    )

    if not isinstance(
        speaker_object,
        dict,
    ):
        return {
            "label": None,
            "canonical": None,
            "base_status": None,
            "boundary_type": None,
        }

    base = speaker_object.get(
        "base",
        {}
    )

    effective = speaker_object.get(
        "effective",
        {}
    )

    if not isinstance(
        base,
        dict,
    ):
        base = {}

    if not isinstance(
        effective,
        dict,
    ):
        effective = {}

    return {
        "label": effective.get(
            "label"
        ),
        "canonical": effective.get(
            "canonical"
        ),
        "base_status": base.get(
            "status"
        ),
        "boundary_type": base.get(
            "boundary_type"
        ),
    }

# -------------------------------------------------------------------
# Full speaker-label profile
# -------------------------------------------------------------------

program_label_counts = {}
program_canonical_counts = {}

print(
    "\n=== OBSERVED SPEAKER LABELS ==="
)

for program_id in target_program_ids:
    records = (
        target_program_records[
            program_id
        ]
    )

    label_counts = Counter()
    canonical_counts = Counter()

    for record in records:
        speaker_fields = (
            get_effective_speaker_fields(
                record
            )
        )

        label_counts[
            speaker_fields[
                "label"
            ]
        ] += 1

        canonical_counts[
            speaker_fields[
                "canonical"
            ]
        ] += 1

    program_label_counts[
        program_id
    ] = label_counts

    program_canonical_counts[
        program_id
    ] = canonical_counts

    print(
        "\nPROGRAM:",
        program_id,
    )

    print(
        "Speaker labels:"
    )

    for label, count in (
        label_counts.most_common()
    ):
        print(
            "-",
            repr(label),
            ":",
            count,
        )

    print(
        "Canonical speakers:"
    )

    for canonical, count in (
        canonical_counts.most_common()
    ):
        print(
            "-",
            repr(canonical),
            ":",
            count,
        )

# -------------------------------------------------------------------
# Season / episode structure
# -------------------------------------------------------------------

print(
    "\n=== PROGRAM SEASON / EPISODE STRUCTURE ==="
)

program_structure_profile = {}

for program_id in target_program_ids:
    records = (
        target_program_records[
            program_id
        ]
    )

    season_counts = Counter()
    episode_counts = Counter()
    source_ids = set()

    for record in records:
        metadata = record.get(
            "source_metadata",
            {}
        )

        season_counts[
            metadata.get(
                "season_number"
            )
        ] += 1

        episode_counts[
            metadata.get(
                "episode_number"
            )
        ] += 1

        source_ids.add(
            record[
                "source_id"
            ]
        )

    program_structure_profile[
        program_id
    ] = {
        "documents": len(
            records
        ),
        "source_ids": len(
            source_ids
        ),
        "season_counts": (
            season_counts
        ),
        "episode_counts": (
            episode_counts
        ),
    }

    print(
        "\nPROGRAM:",
        program_id,
    )

    print(
        "Distinct source IDs:",
        len(
            source_ids
        ),
    )

    print(
        "Season values:",
        dict(
            season_counts
        ),
    )

    print(
        "Distinct non-null episode numbers:",
        len(
            {
                value
                for value
                in episode_counts
                if value is not None
            }
        ),
    )

# -------------------------------------------------------------------
# Asre Shirin participant-label analysis
# -------------------------------------------------------------------

print(
    "\n=== ASRE SHIRIN PARTICIPANT LABEL AUDIT ==="
)

asre_records = (
    target_program_records[
        "program:asre-shirin"
    ]
)

asre_participant_label_counts = Counter()
asre_nonparticipant_labels = Counter()

participant_markers = [
    "شرکت‌کننده",
    "شرکت کننده",
    "یکی از شرکت",
]

for record in asre_records:
    speaker_fields = (
        get_effective_speaker_fields(
            record
        )
    )

    label = speaker_fields[
        "label"
    ]

    if not isinstance(
        label,
        str,
    ):
        continue

    if any(
        marker in label
        for marker
        in participant_markers
    ):
        asre_participant_label_counts[
            label
        ] += 1
    else:
        asre_nonparticipant_labels[
            label
        ] += 1

print(
    "Observed participant-like labels:"
)

for label, count in (
    asre_participant_label_counts.most_common()
):
    print(
        "-",
        repr(label),
        ":",
        count,
    )

print(
    "\nOther explicit labels:"
)

for label, count in (
    asre_nonparticipant_labels.most_common()
):
    print(
        "-",
        repr(label),
        ":",
        count,
    )

print(
    "\nDistinct participant-like labels:",
    len(
        asre_participant_label_counts
    ),
)

# -------------------------------------------------------------------
# Khane-To structure anomaly discovery
# -------------------------------------------------------------------

print(
    "\n=== KHANE-TO TWO-PERSON STRUCTURE AUDIT ==="
)

khane_records = (
    target_program_records[
        "program:khane-to"
    ]
)

khane_label_counts = (
    program_label_counts[
        "program:khane-to"
    ]
)

# Human confirmation says:
# host + Dr. Roudi only.
#
# We do not guess the host identity here.
# We identify the dominant non-primary label
# and separately surface all remaining labels.

khane_primary_label = (
    "دکتر کمیل رودی"
)

khane_nonprimary_counts = Counter(
    {
        label: count
        for label, count
        in khane_label_counts.items()
        if (
            label is not None
            and label
            != khane_primary_label
        )
    }
)

khane_dominant_nonprimary_label = None

if khane_nonprimary_counts:
    khane_dominant_nonprimary_label = (
        khane_nonprimary_counts
        .most_common(
            1
        )[0][0]
    )

print(
    "Primary explicit label:",
    repr(
        khane_primary_label
    ),
)

print(
    "Dominant non-primary label:",
    repr(
        khane_dominant_nonprimary_label
    ),
)

khane_review_labels = Counter()

for label, count in (
    khane_nonprimary_counts.items()
):
    if (
        label
        != khane_dominant_nonprimary_label
    ):
        khane_review_labels[
            label
        ] = count

print(
    "Labels outside primary + dominant non-primary:"
)

if khane_review_labels:
    for label, count in (
        khane_review_labels.most_common()
    ):
        print(
            "-",
            repr(label),
            ":",
            count,
        )
else:
    print(
        "- NONE"
    )

# -------------------------------------------------------------------
# Examples for Khane-To labels requiring review
# -------------------------------------------------------------------

print(
    "\n=== KHANE-TO REVIEW-LABEL EXAMPLES ==="
)

khane_review_examples = defaultdict(
    list
)

for record in khane_records:
    speaker_fields = (
        get_effective_speaker_fields(
            record
        )
    )

    label = speaker_fields[
        "label"
    ]

    if label not in (
        khane_review_labels
    ):
        continue

    if len(
        khane_review_examples[
            label
        ]
    ) < 10:
        khane_review_examples[
            label
        ].append(
            {
                "document_id": (
                    record[
                        "document_id"
                    ]
                ),
                "source_id": (
                    record[
                        "source_id"
                    ]
                ),
                "label": label,
                "base_status": (
                    speaker_fields[
                        "base_status"
                    ]
                ),
                "boundary_type": (
                    speaker_fields[
                        "boundary_type"
                    ]
                ),
                "text": (
                    record[
                        "text"
                    ]
                ),
            }
        )

for label in sorted(
    khane_review_examples,
    key=str,
):
    print(
        "\nLABEL:",
        repr(label),
    )

    for example in (
        khane_review_examples[
            label
        ]
    ):
        print(
            "-",
            example[
                "document_id"
            ],
            "|",
            example[
                "base_status"
            ],
            "|",
            example[
                "boundary_type"
            ],
        )

        print(
            " ",
            example[
                "text"
            ]
            .replace(
                "\n",
                " ",
            )[:450],
        )

# -------------------------------------------------------------------
# Human-confirmed vs observed summary
# -------------------------------------------------------------------

print(
    "\n=== HUMAN-CONFIRMED VS OBSERVED SUMMARY ==="
)

asre_nonnull_seasons = {
    value
    for value
    in program_structure_profile[
        "program:asre-shirin"
    ][
        "season_counts"
    ]
    if value is not None
}

asre_two_seasons_observed = (
    len(
        asre_nonnull_seasons
    )
    == 2
)

khane_two_dominant_roles_observed = (
    khane_primary_label
    in khane_label_counts
    and khane_dominant_nonprimary_label
    is not None
)

print(
    "Asre Shirin human-confirmed seasons:",
    2,
)

print(
    "Asre Shirin observed non-null seasons:",
    sorted(
        asre_nonnull_seasons,
        key=str,
    ),
)

print(
    "Asre Shirin two-season structure observed:",
    asre_two_seasons_observed,
)

print(
    "\nKhane-To human-confirmed structure:",
    "host + Dr. Komeil Roudi",
)

print(
    "Khane-To dominant observed roles:",
    {
        "primary_label": (
            khane_primary_label
        ),
        "dominant_nonprimary_label": (
            khane_dominant_nonprimary_label
        ),
    },
)

print(
    "Khane-To labels requiring interpretation:",
    dict(
        khane_review_labels
    ),
)

# -------------------------------------------------------------------
# Validation
# -------------------------------------------------------------------

target_population_accounted = all(
    len(
        target_program_records[
            program_id
        ]
    )
    > 0
    for program_id
    in target_program_ids
)

speaker_profiles_complete = all(
    sum(
        program_label_counts[
            program_id
        ].values()
    )
    == len(
        target_program_records[
            program_id
        ]
    )
    for program_id
    in target_program_ids
)

structure_profiles_complete = all(
    program_structure_profile[
        program_id
    ][
        "documents"
    ]
    == len(
        target_program_records[
            program_id
        ]
    )
    for program_id
    in target_program_ids
)

human_context_recorded = (
    human_confirmed_program_context[
        "program:khane-to"
    ][
        "structure"
    ]
    == "TWO_PERSON_HOST_AND_DR_ROUDI"
    and human_confirmed_program_context[
        "program:asre-shirin"
    ][
        "season_count"
    ]
    == 2
)

no_dataset_mutation = (
    len(
        quality_records
    )
    == 19456
    and len(
        transcript_records_current
    )
    == 16226
)

print(
    "\nValidation:"
)

print(
    "Target program populations found:",
    "PASS"
    if target_population_accounted
    else "FAIL",
)

print(
    "Speaker-label profiles account for every target segment:",
    "PASS"
    if speaker_profiles_complete
    else "FAIL",
)

print(
    "Program structure profiles complete:",
    "PASS"
    if structure_profiles_complete
    else "FAIL",
)

print(
    "Human-confirmed program context recorded:",
    "PASS"
    if human_context_recorded
    else "FAIL",
)

print(
    "No dataset mutation:",
    "PASS"
    if no_dataset_mutation
    else "FAIL",
)

cell_gate = (
    target_population_accounted
    and speaker_profiles_complete
    and structure_profiles_complete
    and human_context_recorded
    and no_dataset_mutation
)

print(
    "CODE CELL 31 GATE:",
    "PASS"
    if cell_gate
    else "FAIL",
)
```

<h3>طبقه‌بندی Evidenceهای کیفیت بدون امتیازدهی و بدون Quarantine</h3>

<p>
تا این مرحله Dataset از چند جهت مستقل بررسی شده است:
</p>

<ul>
<li>تمام 19,456 Document با JSON Schema معتبر هستند.</li>
<li>هیچ متن خالی یا whitespace-only وجود ندارد.</li>
<li>Locatorهای کامل در هر سه Source یکتا هستند.</li>
<li>Lineage Hash برای تمام رکوردها با <code>raw_content_hash</code> سازگار است.</li>
<li>Rights Evidence برای بسیاری از منابع هنوز باز است.</li>
<li>در Transcript بخشی از Speakerها Primary، بخشی Contextual و بخشی Unresolved هستند.</li>
<li>وجود Exact Duplicate یا Similarity بالا به معنی حذف Source نیست.</li>
<li>برخی Metadata gapها مانند Season در «عصر شیرین» Schema-valid ولی برای Retrieval و تحلیل ساختاری مهم هستند.</li>
</ul>

<h4>هدف این مرحله</h4>

<p>
قبل از ساخت هر Quality Score باید Evidenceها در گروه‌های معنایی جدا قرار گیرند.
این کار مانع می‌شود یک Metadata gap نرم با یک Structural Failure سخت اشتباه گرفته شود.
</p>

<h4>طبقات کیفیت</h4>

<h5>1. Hard Structural Failure</h5>

<p>
مواردی که در صورت وجود می‌توانند مانع عبور یک Document شوند:
</p>

<ul>
<li>JSON Schema invalidity</li>
<li>متن خالی</li>
<li>Locator غیرقابل استفاده یا تکراری در سطح هویت Document</li>
<li>Lineage Hash inconsistency</li>
<li>Missing required identity fields</li>
</ul>

<p>
بر اساس Discoveryهای فعلی انتظار می‌رود این Population صفر باشد.
</p>

<h5>2. Rights Open Gate</h5>

<p>
وضعیت‌هایی مانند:
</p>

<p><code>EVIDENCE_REQUIRED</code></p>

<p>
به معنی Hard Quality Failure نیستند.
این وضعیت نشان می‌دهد Evidence مستقل حقوقی هنوز تکمیل نشده است.
</p>

<h5>3. Attribution / Authority Context</h5>

<p>
در Transcript باید میان موارد زیر تفاوت حفظ شود:
</p>

<ul>
<li>Primary evidence مربوط به دکتر کمیل رودی</li>
<li>مجری، مهمان و Participant به‌عنوان Context</li>
<li>Speaker واقعاً Unresolved</li>
</ul>

<p>
Contextual بودن یک Segment به معنی کیفیت پایین نیست.
</p>

<h5>4. Soft Metadata Gap</h5>

<p>
نمونه‌ها:
</p>

<ul>
<li>Language خالی ولی Schema-valid</li>
<li>Content Origin خالی ولی Schema-valid</li>
<li>Season metadata ثبت‌نشده برای عصر شیرین</li>
<li>Speaker unresolved بدون اجازه برای حدس</li>
<li>Rights evidence تکمیل‌نشده</li>
</ul>

<h5>5. Review Candidate</h5>

<p>
مواردی که Evidence متناقض یا غیرعادی دارند ولی هنوز برای Quarantine کافی نیستند.
برای نمونه در «خانه تو» ساختار انسانی برنامه دو نفره است،
اما 7 Segment با Labelهای «شرکت‌کننده 1/2» دیده شده‌اند.
این موارد فقط برای Review علامت‌گذاری می‌شوند.
</p>

<h5>6. Dedup Relation</h5>

<p>
Exact Duplicate یا Near-Duplicate یک Relation میان اسناد است،
نه لزوماً Quality Failure.
Source Identity و Lineage باید حفظ شوند.
</p>

<h4>اصل مهم</h4>

<p>
در این مرحله هنوز:
</p>

<ul>
<li>هیچ Weight عددی تعریف نمی‌شود.</li>
<li>هیچ Quality Score ساخته نمی‌شود.</li>
<li>هیچ Document حذف نمی‌شود.</li>
<li>هیچ Document Quarantine نمی‌شود.</li>
</ul>

<p>
فقط یک Evidence Matrix برای تصمیم‌گیری مرحله بعد ساخته می‌شود.
</p>

<h4>معیار PASS</h4>

<p>
هر 19,456 Document باید دقیقاً یک‌بار ارزیابی شود،
Hard Failure population باید بر پایه Evidence واقعی محاسبه شود
و Soft Gap / Rights / Authority / Review / Dedup flags باید بدون تغییر Dataset ثبت شوند.
</p>

</div>

```python
# Code Cell 32

from collections import Counter, defaultdict
import hashlib
import json

# -------------------------------------------------------------------
# Independent exact-text duplicate membership
# -------------------------------------------------------------------

text_hash_counts_current = Counter(
    record[
        "text_sha256"
    ]
    for record
    in quality_records
)

exact_duplicate_hashes_current = {
    text_hash
    for text_hash, count
    in text_hash_counts_current.items()
    if count > 1
}

# -------------------------------------------------------------------
# Full locator uniqueness lookup
# -------------------------------------------------------------------

full_locator_counts_by_source = {}

for source_type in [
    "website",
    "book",
    "transcript",
]:
    serialized_locators = []

    for record in quality_records:
        if (
            record[
                "source_type"
            ]
            != source_type
        ):
            continue

        serialized = json.dumps(
            record[
                "locator"
            ],
            ensure_ascii=False,
            sort_keys=True,
            separators=(
                ",",
                ":",
            ),
        )

        serialized_locators.append(
            serialized
        )

    full_locator_counts_by_source[
        source_type
    ] = Counter(
        serialized_locators
    )

# -------------------------------------------------------------------
# Transcript authority lookup from Cell 29
# -------------------------------------------------------------------

transcript_authority_lookup = {
    row[
        "document_id"
    ]: row
    for row in (
        final_transcript_authority_rows
    )
}

# -------------------------------------------------------------------
# Human-confirmed program review lookup
# -------------------------------------------------------------------

khane_to_review_document_ids = set()

for label, examples in (
    khane_review_examples.items()
):
    for example in examples:
        khane_to_review_document_ids.add(
            example[
                "document_id"
            ]
        )

# The examples dictionary only retained up to 10 per label.
# Rebuild the full 7-document population independently.

for record in (
    target_program_records[
        "program:khane-to"
    ]
):
    speaker_fields = (
        get_effective_speaker_fields(
            record
        )
    )

    if speaker_fields[
        "label"
    ] in {
        "شرکت‌کننده 1",
        "شرکت‌کننده 2",
    }:
        khane_to_review_document_ids.add(
            record[
                "document_id"
            ]
        )

# -------------------------------------------------------------------
# Quality evidence classifier
# -------------------------------------------------------------------

quality_evidence_rows = []

for record in quality_records:
    document_id = record[
        "document_id"
    ]

    source_type = record[
        "source_type"
    ]

    text = record[
        "text"
    ]

    locator_serialized = json.dumps(
        record[
            "locator"
        ],
        ensure_ascii=False,
        sort_keys=True,
        separators=(
            ",",
            ":",
        ),
    )

    lineage_hash = (
        record[
            "lineage"
        ].get(
            "raw_source_hash"
        )
    )

    raw_content_hash = (
        record.get(
            "raw_content_hash"
        )
    )

    # ---------------------------------------------------------------
    # Hard structural evidence
    # ---------------------------------------------------------------

    hard_schema_failure = False
    # Code Cell 30 independently established
    # 19,456 / 19,456 schema-valid documents.

    hard_empty_text = (
        not isinstance(
            text,
            str,
        )
        or text.strip()
        == ""
    )

    hard_duplicate_full_locator = (
        full_locator_counts_by_source[
            source_type
        ][
            locator_serialized
        ]
        > 1
    )

    hard_lineage_hash_mismatch = (
        lineage_hash
        != raw_content_hash
    )

    hard_failure_reasons = []

    if hard_schema_failure:
        hard_failure_reasons.append(
            "SCHEMA_INVALID"
        )

    if hard_empty_text:
        hard_failure_reasons.append(
            "EMPTY_TEXT"
        )

    if hard_duplicate_full_locator:
        hard_failure_reasons.append(
            "DUPLICATE_FULL_LOCATOR"
        )

    if hard_lineage_hash_mismatch:
        hard_failure_reasons.append(
            "LINEAGE_HASH_MISMATCH"
        )

    # ---------------------------------------------------------------
    # Rights evidence
    # ---------------------------------------------------------------

    rights = record[
        "rights"
    ]

    rights_status = (
        rights.get(
            "status"
        )
    )

    rights_basis = (
        rights.get(
            "basis"
        )
    )

    rights_evidence_id = (
        rights.get(
            "rights_evidence_id"
        )
    )

    if (
        rights_status
        == "EVIDENCE_REQUIRED"
    ):
        rights_evidence_class = (
            "RIGHTS_EVIDENCE_REQUIRED"
        )

    elif (
        rights_status is None
        and rights_evidence_id
        is None
    ):
        rights_evidence_class = (
            "RIGHTS_STATUS_NOT_ESTABLISHED"
        )

    else:
        rights_evidence_class = (
            "RIGHTS_STATUS_PRESENT"
        )

    # ---------------------------------------------------------------
    # Soft metadata gaps
    # ---------------------------------------------------------------

    soft_metadata_flags = []

    if record.get(
        "language"
    ) is None:
        soft_metadata_flags.append(
            "LANGUAGE_NULL_SCHEMA_VALID"
        )

    if record.get(
        "content_origin"
    ) is None:
        soft_metadata_flags.append(
            "CONTENT_ORIGIN_NULL_SCHEMA_VALID"
        )

    # Human-confirmed two-season structure exists
    # for Asre Shirin, but season metadata is absent
    # in the current normalized dataset.
    if source_type == "transcript":
        metadata = record[
            "source_metadata"
        ]

        program_id = metadata.get(
            "program_id"
        )

        if (
            program_id
            == "program:asre-shirin"
            and metadata.get(
                "season_number"
            )
            is None
        ):
            soft_metadata_flags.append(
                "ASRE_SHIRIN_SEASON_METADATA_MISSING"
            )

    # ---------------------------------------------------------------
    # Authority / attribution evidence
    # ---------------------------------------------------------------

    authority_class = None
    authority_basis = None

    if source_type == "website":
        content_origin = (
            record.get(
                "content_origin"
            )
        )

        if (
            content_origin
            == "primary_author"
        ):
            authority_class = (
                "PRIMARY_DIRECT_WEBSITE"
            )

        elif (
            content_origin
            == "supervised"
        ):
            authority_class = (
                "SECONDARY_SUPERVISED_WEBSITE"
            )

        else:
            authority_class = (
                "WEBSITE_AUTHORITY_UNRESOLVED"
            )

        authority_basis = (
            "WEBSITE_CONTENT_ORIGIN"
        )

    elif source_type == "book":
        attribution = record[
            "attribution"
        ]

        if (
            attribution.get(
                "author"
            )
            == "Dr. Komeil Roudi"
            and attribution.get(
                "knowledge_owner"
            )
            == "Dr. Komeil Roudi"
        ):
            authority_class = (
                "PRIMARY_DIRECT_BOOK"
            )
        else:
            authority_class = (
                "BOOK_AUTHORITY_REVIEW"
            )

        authority_basis = (
            "BOOK_AUTHOR_AND_KNOWLEDGE_OWNER"
        )

    elif source_type == "transcript":
        authority_row = (
            transcript_authority_lookup.get(
                document_id
            )
        )

        if authority_row is None:
            authority_class = (
                "TRANSCRIPT_AUTHORITY_MISSING"
            )
            authority_basis = (
                "NO_AUTHORITY_ROW"
            )
        else:
            authority_class = (
                authority_row[
                    "authority_class"
                ]
            )

            authority_basis = (
                authority_row[
                    "authority_basis"
                ]
            )

            if (
                authority_class
                == "UNRESOLVED_SPEAKER_CONTEXT"
            ):
                soft_metadata_flags.append(
                    "SPEAKER_UNRESOLVED_DO_NOT_INFER"
                )

    # ---------------------------------------------------------------
    # Review candidates
    # ---------------------------------------------------------------

    review_flags = []

    if (
        document_id
        in khane_to_review_document_ids
    ):
        review_flags.append(
            "KHANE_TO_UNEXPECTED_PARTICIPANT_LABEL"
        )

    # ---------------------------------------------------------------
    # Dedup relation
    # ---------------------------------------------------------------

    exact_duplicate_member = (
        record[
            "text_sha256"
        ]
        in exact_duplicate_hashes_current
    )

    if exact_duplicate_member:
        dedup_relation_class = (
            "EXACT_TEXT_DUPLICATE_MEMBER"
        )
    else:
        dedup_relation_class = (
            "NO_EXACT_TEXT_DUPLICATE"
        )

    # ---------------------------------------------------------------
    # Record
    # ---------------------------------------------------------------

    quality_evidence_rows.append(
        {
            "document_id": (
                document_id
            ),
            "source_type": (
                source_type
            ),
            "source_id": (
                record[
                    "source_id"
                ]
            ),
            "hard_failure": bool(
                hard_failure_reasons
            ),
            "hard_failure_reasons": (
                hard_failure_reasons
            ),
            "rights_evidence_class": (
                rights_evidence_class
            ),
            "rights_status": (
                rights_status
            ),
            "rights_basis": (
                rights_basis
            ),
            "soft_metadata_flags": (
                soft_metadata_flags
            ),
            "authority_class": (
                authority_class
            ),
            "authority_basis": (
                authority_basis
            ),
            "review_flags": (
                review_flags
            ),
            "dedup_relation_class": (
                dedup_relation_class
            ),
        }
    )

# -------------------------------------------------------------------
# Global evidence profile
# -------------------------------------------------------------------

print(
    "=== QUALITY EVIDENCE CLASSIFICATION ==="
)

print(
    "Documents:",
    len(
        quality_evidence_rows
    ),
)

hard_failure_rows = [
    row
    for row in quality_evidence_rows
    if row[
        "hard_failure"
    ]
]

print(
    "\nHard structural failure documents:",
    len(
        hard_failure_rows
    ),
)

hard_failure_reason_counts = Counter(
    reason
    for row in hard_failure_rows
    for reason in row[
        "hard_failure_reasons"
    ]
)

print(
    "Hard failure reasons:",
    dict(
        hard_failure_reason_counts
    ),
)

# -------------------------------------------------------------------
# Rights profile
# -------------------------------------------------------------------

rights_evidence_counts = Counter(
    row[
        "rights_evidence_class"
    ]
    for row in quality_evidence_rows
)

print(
    "\n=== RIGHTS EVIDENCE PROFILE ==="
)

for value, count in (
    rights_evidence_counts.most_common()
):
    print(
        "-",
        value,
        ":",
        count,
    )

# -------------------------------------------------------------------
# Soft metadata profile
# -------------------------------------------------------------------

soft_flag_counts = Counter(
    flag
    for row in quality_evidence_rows
    for flag in row[
        "soft_metadata_flags"
    ]
)

documents_with_soft_flags = sum(
    bool(
        row[
            "soft_metadata_flags"
        ]
    )
    for row in quality_evidence_rows
)

print(
    "\n=== SOFT METADATA GAP PROFILE ==="
)

print(
    "Documents with >=1 soft metadata flag:",
    documents_with_soft_flags,
)

for flag, count in (
    soft_flag_counts.most_common()
):
    print(
        "-",
        flag,
        ":",
        count,
    )

# -------------------------------------------------------------------
# Authority profile
# -------------------------------------------------------------------

authority_counts = Counter(
    row[
        "authority_class"
    ]
    for row in quality_evidence_rows
)

print(
    "\n=== AUTHORITY / ATTRIBUTION PROFILE ==="
)

for value, count in (
    authority_counts.most_common()
):
    print(
        "-",
        value,
        ":",
        count,
    )

# -------------------------------------------------------------------
# Review candidates
# -------------------------------------------------------------------

review_flag_counts = Counter(
    flag
    for row in quality_evidence_rows
    for flag in row[
        "review_flags"
    ]
)

review_document_count = sum(
    bool(
        row[
            "review_flags"
        ]
    )
    for row in quality_evidence_rows
)

print(
    "\n=== REVIEW CANDIDATE PROFILE ==="
)

print(
    "Documents requiring review:",
    review_document_count,
)

for flag, count in (
    review_flag_counts.most_common()
):
    print(
        "-",
        flag,
        ":",
        count,
    )

# -------------------------------------------------------------------
# Dedup relation profile
# -------------------------------------------------------------------

dedup_relation_counts = Counter(
    row[
        "dedup_relation_class"
    ]
    for row in quality_evidence_rows
)

print(
    "\n=== EXACT DEDUP RELATION PROFILE ==="
)

for value, count in (
    dedup_relation_counts.most_common()
):
    print(
        "-",
        value,
        ":",
        count,
    )

# -------------------------------------------------------------------
# Source-level summary
# -------------------------------------------------------------------

print(
    "\n=== SOURCE-LEVEL QUALITY EVIDENCE SUMMARY ==="
)

for source_type in [
    "website",
    "book",
    "transcript",
]:
    rows = [
        row
        for row in quality_evidence_rows
        if row[
            "source_type"
        ]
        == source_type
    ]

    print(
        "\nSOURCE:",
        source_type,
    )

    print(
        "Documents:",
        len(
            rows
        ),
    )

    print(
        "Hard failures:",
        sum(
            row[
                "hard_failure"
            ]
            for row in rows
        ),
    )

    print(
        "With soft metadata flags:",
        sum(
            bool(
                row[
                    "soft_metadata_flags"
                ]
            )
            for row in rows
        ),
    )

    print(
        "With review flags:",
        sum(
            bool(
                row[
                    "review_flags"
                ]
            )
            for row in rows
        ),
    )

    print(
        "Exact duplicate members:",
        sum(
            row[
                "dedup_relation_class"
            ]
            == "EXACT_TEXT_DUPLICATE_MEMBER"
            for row in rows
        ),
    )

    print(
        "Rights evidence:",
        dict(
            Counter(
                row[
                    "rights_evidence_class"
                ]
                for row in rows
            )
        ),
    )

# -------------------------------------------------------------------
# Independent consistency checks
# -------------------------------------------------------------------

exact_duplicate_member_count = sum(
    row[
        "dedup_relation_class"
    ]
    == "EXACT_TEXT_DUPLICATE_MEMBER"
    for row in quality_evidence_rows
)

expected_exact_duplicate_member_count = sum(
    count
    for text_hash, count
    in text_hash_counts_current.items()
    if count > 1
)

population_accounted = (
    len(
        quality_evidence_rows
    )
    == 19456
)

hard_failure_expected_zero = (
    len(
        hard_failure_rows
    )
    == 0
)

exact_duplicate_population_reproduced = (
    exact_duplicate_member_count
    == expected_exact_duplicate_member_count
    == 931
)

khane_review_population_reproduced = (
    review_flag_counts.get(
        "KHANE_TO_UNEXPECTED_PARTICIPANT_LABEL",
        0,
    )
    == 7
)

asre_season_gap_reproduced = (
    soft_flag_counts.get(
        "ASRE_SHIRIN_SEASON_METADATA_MISSING",
        0,
    )
    == 902
)

transcript_authority_population_present = (
    sum(
        count
        for authority_class, count
        in authority_counts.items()
        if (
            authority_class.startswith(
                "PRIMARY_"
            )
            or authority_class
            in {
                "SECONDARY_HUMAN_CONFIRMED_OVERLAY",
                "NONPRIMARY_EXPLICIT_CONTEXT",
                "UNRESOLVED_SPEAKER_CONTEXT",
            }
        )
    )
    >= 16226
)

no_input_mutation = (
    len(
        quality_records
    )
    == 19456
    and len(
        transcript_records_current
    )
    == 16226
)

print(
    "\nValidation:"
)

print(
    "All documents classified:",
    "PASS"
    if population_accounted
    else "FAIL",
)

print(
    "Hard structural failure population remains zero:",
    "PASS"
    if hard_failure_expected_zero
    else "FAIL",
)

print(
    "Exact duplicate member population reproduced (931):",
    "PASS"
    if exact_duplicate_population_reproduced
    else "FAIL",
)

print(
    "Khane-To 7 review candidates reproduced:",
    "PASS"
    if khane_review_population_reproduced
    else "FAIL",
)

print(
    "Asre Shirin 902 missing-season flags reproduced:",
    "PASS"
    if asre_season_gap_reproduced
    else "FAIL",
)

print(
    "Transcript authority evidence represented:",
    "PASS"
    if transcript_authority_population_present
    else "FAIL",
)

print(
    "Input dataset preserved without mutation:",
    "PASS"
    if no_input_mutation
    else "FAIL",
)

cell_gate = (
    population_accounted
    and hard_failure_expected_zero
    and exact_duplicate_population_reproduced
    and khane_review_population_reproduced
    and asre_season_gap_reproduced
    and transcript_authority_population_present
    and no_input_mutation
)

print(
    "CODE CELL 32 GATE:",
    "PASS"
    if cell_gate
    else "FAIL",
)
```

```python
# Code Cell 33

from pathlib import Path
from collections import Counter, defaultdict
import re

# -------------------------------------------------------------------
# Official roadmap candidates already discovered in repository
# -------------------------------------------------------------------

roadmap_paths = [
    (
        repo_root
        / "data/Road_Maps/"
          "RAG_finance_phase2_modular_notebooks_execution_plan.md"
    ),
    (
        repo_root
        / "data/Road_Maps/"
          "RAG_finance_phase2_operational_roadmap.md"
    ),
    (
        repo_root
        / "data/Road_Maps/"
          "RAG_finance_master_roadmap.md"
    ),
]

print(
    "=== NOTEBOOK 09 ROADMAP CONTRACT DISCOVERY ==="
)

for path in roadmap_paths:
    print(
        "-",
        path.relative_to(
            repo_root
        ),
        "| exists:",
        path.exists(),
    )

# -------------------------------------------------------------------
# Read roadmaps
# -------------------------------------------------------------------

roadmap_texts = {}

for path in roadmap_paths:
    if not path.exists():
        continue

    roadmap_texts[
        str(
            path.relative_to(
                repo_root
            )
        )
    ] = path.read_text(
        encoding="utf-8"
    )

print(
    "\nRoadmaps loaded:",
    len(
        roadmap_texts
    ),
)

# -------------------------------------------------------------------
# Search terms
# -------------------------------------------------------------------

contract_terms = [
    "09_deduplication_and_quality",
    "Notebook 09",
    "Notebook 9",
    "deduplication",
    "dedup",
    "near-duplicate",
    "near duplicate",
    "cross-source",
    "canonical relation",
    "canonical",
    "quality score",
    "quality scoring",
    "quality",
    "quarantine",
    "accepted",
    "release",
    "report",
    "artifact",
]

# -------------------------------------------------------------------
# Extract focused contexts
# -------------------------------------------------------------------

roadmap_contract_hits = []

for relative_path, text in (
    roadmap_texts.items()
):
    lines = text.splitlines()

    for line_index, line in enumerate(
        lines
    ):
        lower_line = line.lower()

        matched_terms = [
            term
            for term in contract_terms
            if term.lower()
            in lower_line
        ]

        if not matched_terms:
            continue

        # Give larger context around explicit Notebook 09 references.
        if any(
            marker.lower()
            in lower_line
            for marker in [
                "09_deduplication_and_quality",
                "notebook 09",
                "notebook 9",
            ]
        ):
            before = 8
            after = 35
        else:
            before = 3
            after = 5

        start = max(
            0,
            line_index - before,
        )

        end = min(
            len(lines),
            line_index + after + 1,
        )

        context = "\n".join(
            f"{idx + 1}: {lines[idx]}"
            for idx in range(
                start,
                end,
            )
        )

        roadmap_contract_hits.append(
            {
                "path": (
                    relative_path
                ),
                "anchor_line": (
                    line_index + 1
                ),
                "matched_terms": sorted(
                    set(
                        matched_terms
                    )
                ),
                "context": (
                    context
                ),
            }
        )

# -------------------------------------------------------------------
# Deduplicate overlapping contexts
# -------------------------------------------------------------------

deduplicated_hits = []

seen_keys = set()

for hit in roadmap_contract_hits:
    normalized = re.sub(
        r"\s+",
        " ",
        hit[
            "context"
        ],
    ).strip()

    key = (
        hit[
            "path"
        ],
        normalized,
    )

    if key in seen_keys:
        continue

    seen_keys.add(
        key
    )

    deduplicated_hits.append(
        hit
    )

print(
    "\n=== ROADMAP CONTRACT CONTEXTS ==="
)

print(
    "Context hits:",
    len(
        deduplicated_hits
    ),
)

for index, hit in enumerate(
    deduplicated_hits[
        :120
    ],
    start=1,
):
    print(
        "\n"
        + "=" * 110
    )

    print(
        "HIT:",
        index,
    )

    print(
        "File:",
        hit[
            "path"
        ],
    )

    print(
        "Anchor line:",
        hit[
            "anchor_line"
        ],
    )

    print(
        "Matched terms:",
        hit[
            "matched_terms"
        ],
    )

    print(
        hit[
            "context"
        ]
    )

# -------------------------------------------------------------------
# Focused extraction of Notebook 09 sections
# -------------------------------------------------------------------

print(
    "\n=== FOCUSED NOTEBOOK 09 SECTIONS ==="
)

notebook09_sections = []

for relative_path, text in (
    roadmap_texts.items()
):
    lines = text.splitlines()

    for line_index, line in enumerate(
        lines
    ):
        lower_line = line.lower()

        if not any(
            marker in lower_line
            for marker in [
                "09_deduplication_and_quality",
                "notebook 09",
                "notebook 9",
            ]
        ):
            continue

        start = max(
            0,
            line_index - 5,
        )

        end = min(
            len(lines),
            line_index + 80,
        )

        section = "\n".join(
            f"{idx + 1}: {lines[idx]}"
            for idx in range(
                start,
                end,
            )
        )

        notebook09_sections.append(
            {
                "path": (
                    relative_path
                ),
                "anchor_line": (
                    line_index + 1
                ),
                "section": (
                    section
                ),
            }
        )

for index, item in enumerate(
    notebook09_sections,
    start=1,
):
    print(
        "\n"
        + "#" * 110
    )

    print(
        "NOTEBOOK 09 SECTION:",
        index,
    )

    print(
        "File:",
        item[
            "path"
        ],
    )

    print(
        "Anchor line:",
        item[
            "anchor_line"
        ],
    )

    print(
        item[
            "section"
        ]
    )

# -------------------------------------------------------------------
# Discover explicit artifact/path mentions related to NB09 concepts
# -------------------------------------------------------------------

artifact_candidate_lines = []

artifact_keywords = [
    "dedup",
    "quality",
    "quarantine",
    "accepted",
    "canonical",
]

path_markers = [
    "data/",
    "reports/",
    ".json",
    ".jsonl",
    ".parquet",
    ".csv",
]

for relative_path, text in (
    roadmap_texts.items()
):
    for line_number, line in enumerate(
        text.splitlines(),
        start=1,
    ):
        lower_line = line.lower()

        concept_match = any(
            keyword
            in lower_line
            for keyword
            in artifact_keywords
        )

        path_match = any(
            marker
            in line
            for marker
            in path_markers
        )

        if (
            concept_match
            and path_match
        ):
            artifact_candidate_lines.append(
                {
                    "path": (
                        relative_path
                    ),
                    "line_number": (
                        line_number
                    ),
                    "line": (
                        line
                    ),
                }
            )

print(
    "\n=== EXPLICIT ARTIFACT / PATH CANDIDATES ==="
)

print(
    "Candidate lines:",
    len(
        artifact_candidate_lines
    ),
)

for item in artifact_candidate_lines[
    :120
]:
    print(
        "-",
        item[
            "path"
        ],
        ":",
        item[
            "line_number"
        ],
    )

    print(
        " ",
        item[
            "line"
        ],
    )

# -------------------------------------------------------------------
# Contract-term coverage by roadmap
# -------------------------------------------------------------------

print(
    "\n=== CONTRACT TERM COVERAGE BY ROADMAP ==="
)

for relative_path, text in (
    roadmap_texts.items()
):
    lower_text = text.lower()

    present_terms = [
        term
        for term
        in contract_terms
        if term.lower()
        in lower_text
    ]

    print(
        "\nFILE:",
        relative_path,
    )

    print(
        "Terms:",
        present_terms,
    )

# -------------------------------------------------------------------
# Validation
# -------------------------------------------------------------------

all_roadmaps_present = all(
    path.exists()
    for path
    in roadmap_paths
)

all_roadmaps_loaded = (
    len(
        roadmap_texts
    )
    == len(
        roadmap_paths
    )
)

notebook09_contract_found = (
    len(
        notebook09_sections
    )
    > 0
)

quality_contract_found = any(
    "quality"
    in text.lower()
    for text in (
        roadmap_texts.values()
    )
)

quarantine_contract_found = any(
    "quarantine"
    in text.lower()
    for text in (
        roadmap_texts.values()
    )
)

dedup_contract_found = any(
    "dedup"
    in text.lower()
    for text in (
        roadmap_texts.values()
    )
)

print(
    "\nValidation:"
)

print(
    "All official roadmap files present:",
    "PASS"
    if all_roadmaps_present
    else "FAIL",
)

print(
    "All official roadmaps loaded:",
    "PASS"
    if all_roadmaps_loaded
    else "FAIL",
)

print(
    "Notebook 09 contract discovered:",
    "PASS"
    if notebook09_contract_found
    else "FAIL",
)

print(
    "Quality contract evidence discovered:",
    "PASS"
    if quality_contract_found
    else "FAIL",
)

print(
    "Quarantine contract evidence discovered:",
    "PASS"
    if quarantine_contract_found
    else "FAIL",
)

print(
    "Dedup contract evidence discovered:",
    "PASS"
    if dedup_contract_found
    else "FAIL",
)

cell_gate = (
    all_roadmaps_present
    and all_roadmaps_loaded
    and notebook09_contract_found
    and quality_contract_found
    and quarantine_contract_found
    and dedup_contract_found
)

print(
    "CODE CELL 33 GATE:",
    "PASS"
    if cell_gate
    else "FAIL",
)
```

```python
# Code Cell 34

from pathlib import Path
from collections import Counter, defaultdict
import re
import json

# -------------------------------------------------------------------
# High-signal files discovered in prior cells
# -------------------------------------------------------------------

rights_policy_paths = [
    repo_root
    / "data/Road_Maps/"
      "RAG_finance_phase2_operational_roadmap.md",

    repo_root
    / "data/Road_Maps/"
      "RAG_finance_phase2_modular_notebooks_execution_plan.md",

    repo_root
    / "reports/phase2/"
      "book_inventory_acceptance_v1.json",

    repo_root
    / "reports/phase2/"
      "transcript_ingestion_acceptance_v1.json",

    repo_root
    / "reports/phase2/"
      "transcript_inventory_acceptance_v1.json",

    repo_root
    / "reports/phase2/"
      "unified_schema_lineage_acceptance_v1.json",

    repo_root
    / "reports/phase2/"
      "persian_normalization_acceptance_v1.json",

    repo_root
    / "notebooks/phase2/"
      "03_book_inventory.ipynb",

    repo_root
    / "notebooks/phase2/"
      "05_transcript_inventory.ipynb",

    repo_root
    / "notebooks/phase2/"
      "06_transcript_ingestion.ipynb",

    repo_root
    / "notebooks/phase2/"
      "07_unified_schema_and_lineage.ipynb",

    repo_root
    / "notebooks/phase2/"
      "08_persian_normalization.ipynb",
]

print(
    "=== RIGHTS PROCESSING/RELEASE POLICY AUDIT ==="
)

existing_rights_policy_paths = []

for path in rights_policy_paths:
    exists = path.exists()

    print(
        "-",
        path.relative_to(
            repo_root
        ),
        "| exists:",
        exists,
    )

    if exists:
        existing_rights_policy_paths.append(
            path
        )

# -------------------------------------------------------------------
# Focused policy terms
# -------------------------------------------------------------------

rights_terms = [
    "EVIDENCE_REQUIRED",
    "rights_status",
    "rights evidence",
    "rights_evidence",
    "accepted for ingestion",
    "ACCEPTED_FOR_INGESTION",
    "acceptance",
    "accepted",
    "release",
    "release eligibility",
    "release-eligible",
    "quarantine",
    "quarantined",
    "canonical dataset",
    "open gate",
    "open_gates",
    "known limitation",
    "known_limitations",
    "repository presence",
    "مدرک حقوق",
    "فاقد مدرک",
    "بهره‌برداری",
]

# -------------------------------------------------------------------
# Read all high-signal files and extract contexts
# -------------------------------------------------------------------

rights_context_hits = []

for path in existing_rights_policy_paths:
    try:
        text = path.read_text(
            encoding="utf-8"
        )
    except (
        UnicodeDecodeError,
        OSError,
    ):
        continue

    lines = text.splitlines()

    relative_path = str(
        path.relative_to(
            repo_root
        )
    )

    for line_index, line in enumerate(
        lines
    ):
        lower_line = line.lower()

        matched_terms = [
            term
            for term in rights_terms
            if term.lower()
            in lower_line
        ]

        if not matched_terms:
            continue

        start = max(
            0,
            line_index - 4,
        )

        end = min(
            len(lines),
            line_index + 5,
        )

        context = "\n".join(
            f"{idx + 1}: {lines[idx]}"
            for idx in range(
                start,
                end,
            )
        )

        rights_context_hits.append(
            {
                "path": (
                    relative_path
                ),
                "anchor_line": (
                    line_index + 1
                ),
                "matched_terms": sorted(
                    set(
                        matched_terms
                    )
                ),
                "context": (
                    context
                ),
            }
        )

# -------------------------------------------------------------------
# Deduplicate overlapping contexts
# -------------------------------------------------------------------

deduped_rights_hits = []

seen_contexts = set()

for hit in rights_context_hits:
    normalized_context = re.sub(
        r"\s+",
        " ",
        hit[
            "context"
        ],
    ).strip()

    key = (
        hit[
            "path"
        ],
        normalized_context,
    )

    if key in seen_contexts:
        continue

    seen_contexts.add(
        key
    )

    deduped_rights_hits.append(
        hit
    )

print(
    "\nRights-related context hits:",
    len(
        deduped_rights_hits
    ),
)

# -------------------------------------------------------------------
# Score contexts for release/processing relevance
# -------------------------------------------------------------------

def classify_policy_context(
    context,
):
    lower_context = (
        context.lower()
    )

    signals = []

    processing_signals = [
        "accepted_for_ingestion",
        "accepted for ingestion",
        "ingestion",
        "processing",
    ]

    release_signals = [
        "release",
        "canonical dataset",
    ]

    blocking_signals = [
        "quarantine",
        "فاقد مدرک",
        "evidence_required",
        "rights evidence",
        "rights_evidence",
    ]

    if any(
        signal
        in lower_context
        for signal
        in processing_signals
    ):
        signals.append(
            "PROCESSING_CONTEXT"
        )

    if any(
        signal
        in lower_context
        for signal
        in release_signals
    ):
        signals.append(
            "RELEASE_CONTEXT"
        )

    if any(
        signal
        in lower_context
        for signal
        in blocking_signals
    ):
        signals.append(
            "RIGHTS_GATE_CONTEXT"
        )

    return signals

for hit in deduped_rights_hits:
    hit[
        "context_classes"
    ] = classify_policy_context(
        hit[
            "context"
        ]
    )

# -------------------------------------------------------------------
# Print highest-signal contexts
# -------------------------------------------------------------------

priority_hits = sorted(
    deduped_rights_hits,
    key=lambda hit: (
        -len(
            hit[
                "context_classes"
            ]
        ),
        -len(
            hit[
                "matched_terms"
            ]
        ),
        hit[
            "path"
        ],
        hit[
            "anchor_line"
        ],
    ),
)

print(
    "\n=== HIGHEST-SIGNAL RIGHTS POLICY CONTEXTS ==="
)

for index, hit in enumerate(
    priority_hits[
        :120
    ],
    start=1,
):
    print(
        "\n"
        + "=" * 110
    )

    print(
        "HIT:",
        index,
    )

    print(
        "File:",
        hit[
            "path"
        ],
    )

    print(
        "Anchor:",
        hit[
            "anchor_line"
        ],
    )

    print(
        "Context classes:",
        hit[
            "context_classes"
        ],
    )

    print(
        "Terms:",
        hit[
            "matched_terms"
        ],
    )

    print(
        hit[
            "context"
        ]
    )

# -------------------------------------------------------------------
# Focused explicit statements
# -------------------------------------------------------------------

explicit_statement_patterns = {
    "PROCESSING_ALLOWED": [
        r"ACCEPTED_FOR_INGESTION",
        r"accepted for ingestion",
        r"acceptance_decision.{0,80}PASS",
        r"acceptance_status.{0,80}PASS",
    ],
    "RELEASE_BLOCKED_BY_RIGHTS": [
        r"فاقد مدرک.{0,120}Quarantine",
        r"Quarantine.{0,120}rights",
        r"rights.{0,120}Quarantine",
        r"حقوق.{0,120}Quarantine",
    ],
    "RELEASE_RIGHTS_REQUIREMENT": [
        r"Release.{0,160}rights_status",
        r"rights_status.{0,160}Release",
        r"Release.{0,160}Rights",
        r"Rights.{0,160}Release",
    ],
    "RIGHTS_OPEN_GATE": [
        r"EVIDENCE_REQUIRED",
        r"open_gates",
        r"open gate",
        r"known_limitations",
    ],
}

explicit_policy_evidence = defaultdict(
    list
)

for hit in deduped_rights_hits:
    compact_context = re.sub(
        r"\s+",
        " ",
        hit[
            "context"
        ],
    )

    for category, patterns in (
        explicit_statement_patterns.items()
    ):
        for pattern in patterns:
            if re.search(
                pattern,
                compact_context,
                flags=(
                    re.IGNORECASE
                ),
            ):
                explicit_policy_evidence[
                    category
                ].append(
                    hit
                )
                break

print(
    "\n=== EXPLICIT POLICY EVIDENCE COUNTS ==="
)

for category in [
    "PROCESSING_ALLOWED",
    "RELEASE_BLOCKED_BY_RIGHTS",
    "RELEASE_RIGHTS_REQUIREMENT",
    "RIGHTS_OPEN_GATE",
]:
    print(
        "-",
        category,
        ":",
        len(
            explicit_policy_evidence.get(
                category,
                []
            )
        ),
    )

# -------------------------------------------------------------------
# Summarize unique files supporting each policy category
# -------------------------------------------------------------------

print(
    "\n=== POLICY EVIDENCE FILES ==="
)

for category in [
    "PROCESSING_ALLOWED",
    "RELEASE_BLOCKED_BY_RIGHTS",
    "RELEASE_RIGHTS_REQUIREMENT",
    "RIGHTS_OPEN_GATE",
]:
    files = sorted(
        {
            hit[
                "path"
            ]
            for hit
            in explicit_policy_evidence.get(
                category,
                []
            )
        }
    )

    print(
        "\nCATEGORY:",
        category,
    )

    if not files:
        print(
            "- NONE"
        )

    for file_path in files:
        print(
            "-",
            file_path,
        )

# -------------------------------------------------------------------
# Current dataset rights population
# -------------------------------------------------------------------

print(
    "\n=== CURRENT DATASET RIGHTS POPULATION ==="
)

rights_status_by_source = defaultdict(
    Counter
)

rights_basis_by_source = defaultdict(
    Counter
)

for record in quality_records:
    source_type = record[
        "source_type"
    ]

    rights = record[
        "rights"
    ]

    rights_status_by_source[
        source_type
    ][
        rights.get(
            "status"
        )
    ] += 1

    rights_basis_by_source[
        source_type
    ][
        rights.get(
            "basis"
        )
    ] += 1

for source_type in [
    "website",
    "book",
    "transcript",
]:
    print(
        "\nSOURCE:",
        source_type,
    )

    print(
        "rights.status:",
        dict(
            rights_status_by_source[
                source_type
            ]
        ),
    )

    print(
        "rights.basis:",
        dict(
            rights_basis_by_source[
                source_type
            ]
        ),
    )

# -------------------------------------------------------------------
# Decision support — evidence only
# -------------------------------------------------------------------

processing_allowed_evidence = (
    len(
        explicit_policy_evidence[
            "PROCESSING_ALLOWED"
        ]
    )
    > 0
)

release_blocking_rights_evidence = (
    len(
        explicit_policy_evidence[
            "RELEASE_BLOCKED_BY_RIGHTS"
        ]
    )
    > 0
)

release_rights_requirement_evidence = (
    len(
        explicit_policy_evidence[
            "RELEASE_RIGHTS_REQUIREMENT"
        ]
    )
    > 0
)

rights_open_gate_evidence = (
    len(
        explicit_policy_evidence[
            "RIGHTS_OPEN_GATE"
        ]
    )
    > 0
)

print(
    "\n=== POLICY INTERPRETATION SUPPORT ==="
)

print(
    "Evidence that processing/ingestion can proceed:",
    processing_allowed_evidence,
)

print(
    "Evidence that missing rights can block Release:",
    release_blocking_rights_evidence,
)

print(
    "Evidence that Release requires rights metadata/status:",
    release_rights_requirement_evidence,
)

print(
    "Evidence that EVIDENCE_REQUIRED is an open gate/limitation:",
    rights_open_gate_evidence,
)

if (
    processing_allowed_evidence
    and release_blocking_rights_evidence
):
    repository_policy_interpretation = (
        "PROCESSING_ALLOWED_RELEASE_RIGHTS_GATED"
    )

elif (
    processing_allowed_evidence
    and not release_blocking_rights_evidence
):
    repository_policy_interpretation = (
        "PROCESSING_ALLOWED_RELEASE_POLICY_AMBIGUOUS"
    )

else:
    repository_policy_interpretation = (
        "POLICY_AMBIGUOUS"
    )

print(
    "Repository policy interpretation:",
    repository_policy_interpretation,
)

# -------------------------------------------------------------------
# Validation
# -------------------------------------------------------------------

all_high_signal_files_present = (
    len(
        existing_rights_policy_paths
    )
    == len(
        rights_policy_paths
    )
)

contexts_found = (
    len(
        deduped_rights_hits
    )
    > 0
)

processing_policy_found = (
    processing_allowed_evidence
)

release_policy_found = (
    release_blocking_rights_evidence
    or release_rights_requirement_evidence
)

dataset_rights_population_accounted = (
    sum(
        sum(
            counter.values()
        )
        for counter
        in rights_status_by_source.values()
    )
    == 19456
)

no_dataset_mutation = (
    len(
        quality_records
    )
    == 19456
)

print(
    "\nValidation:"
)

print(
    "All high-signal rights-policy files present:",
    "PASS"
    if all_high_signal_files_present
    else "FAIL",
)

print(
    "Rights-policy contexts found:",
    "PASS"
    if contexts_found
    else "FAIL",
)

print(
    "Processing/Ingestion policy evidence found:",
    "PASS"
    if processing_policy_found
    else "FAIL",
)

print(
    "Release rights-gate evidence found:",
    "PASS"
    if release_policy_found
    else "FAIL",
)

print(
    "Current dataset rights population accounted:",
    "PASS"
    if dataset_rights_population_accounted
    else "FAIL",
)

print(
    "No dataset mutation:",
    "PASS"
    if no_dataset_mutation
    else "FAIL",
)

cell_gate = (
    all_high_signal_files_present
    and contexts_found
    and processing_policy_found
    and release_policy_found
    and dataset_rights_population_accounted
    and no_dataset_mutation
)

print(
    "CODE CELL 34 GATE:",
    "PASS"
    if cell_gate
    else "FAIL",
)
```

```python
# Code Cell 35

from pathlib import Path
from collections import Counter, defaultdict
import re

print(
    "=== NOTEBOOK 09 ARTIFACT NAMING DISCOVERY ==="
)

# -------------------------------------------------------------------
# Relevant repository directories
# -------------------------------------------------------------------

processed_dir = (
    repo_root
    / "data/phase2/processed"
)

interim_dir = (
    repo_root
    / "data/phase2/interim"
)

reports_dir = (
    repo_root
    / "reports/phase2"
)

checkpoint_dir = (
    repo_root
    / "artifacts/phase2/checkpoints"
)

required_discovery_directories = [
    processed_dir,
    interim_dir,
    reports_dir,
]

optional_discovery_directories = [
    checkpoint_dir,
]

all_discovery_directories = (
    required_discovery_directories
    + optional_discovery_directories
)

existing_artifact_directories = []

for directory in all_discovery_directories:
    exists = directory.exists()

    role = (
        "REQUIRED_DISCOVERY_ROOT"
        if directory
        in required_discovery_directories
        else "OPTIONAL_CONVENTION_ROOT"
    )

    print(
        "-",
        directory.relative_to(
            repo_root
        ),
        "| exists:",
        exists,
        "| role:",
        role,
    )

    if exists:
        existing_artifact_directories.append(
            directory
        )

checkpoint_directory_exists = (
    checkpoint_dir.exists()
)

# -------------------------------------------------------------------
# List files recursively
# -------------------------------------------------------------------

artifact_files = []

for directory in (
    existing_artifact_directories
):
    for path in sorted(
        directory.rglob(
            "*"
        )
    ):
        if not path.is_file():
            continue

        artifact_files.append(
            path
        )

print(
    "\nArtifact files discovered:",
    len(
        artifact_files
    ),
)

# -------------------------------------------------------------------
# High-signal Phase 2 files
# -------------------------------------------------------------------

high_signal_terms = [
    "acceptance",
    "checkpoint",
    "report",
    "processed",
    "normalized",
    "unified",
    "quality",
    "dedup",
    "duplicate",
    "quarantine",
    "accepted",
    "release",
    "lineage",
    "speaker",
    "book",
    "transcript",
    "website",
]

high_signal_files = []

for path in artifact_files:
    relative_path = str(
        path.relative_to(
            repo_root
        )
    )

    searchable = (
        relative_path.lower()
    )

    matched_terms = [
        term
        for term
        in high_signal_terms
        if term
        in searchable
    ]

    if matched_terms:
        high_signal_files.append(
            {
                "path": path,
                "relative_path": (
                    relative_path
                ),
                "matched_terms": (
                    matched_terms
                ),
            }
        )

print(
    "\n=== HIGH-SIGNAL ARTIFACT FILES ==="
)

for item in high_signal_files:
    path = item[
        "path"
    ]

    print(
        "-",
        item[
            "relative_path"
        ],
        "| bytes:",
        path.stat().st_size,
        "| terms:",
        item[
            "matched_terms"
        ],
    )

# -------------------------------------------------------------------
# Group by parent directory
# -------------------------------------------------------------------

print(
    "\n=== FILES BY ARTIFACT DIRECTORY ==="
)

files_by_parent = defaultdict(
    list
)

for path in artifact_files:
    relative_parent = str(
        path.parent.relative_to(
            repo_root
        )
    )

    files_by_parent[
        relative_parent
    ].append(
        path.name
    )

for parent in sorted(
    files_by_parent
):
    print(
        "\nDIR:",
        parent,
    )

    for name in sorted(
        files_by_parent[
            parent
        ]
    ):
        print(
            "-",
            name,
        )

# -------------------------------------------------------------------
# Discover version suffix conventions
# -------------------------------------------------------------------

version_pattern = re.compile(
    r"_v(\d+)(?=\.[^.]+$)"
)

versioned_files = []

for path in artifact_files:
    match = version_pattern.search(
        path.name
    )

    if match:
        versioned_files.append(
            {
                "relative_path": str(
                    path.relative_to(
                        repo_root
                    )
                ),
                "version": int(
                    match.group(
                        1
                    )
                ),
            }
        )

print(
    "\n=== VERSION SUFFIX CONVENTION ==="
)

print(
    "Versioned files:",
    len(
        versioned_files
    ),
)

version_counts = Counter(
    item[
        "version"
    ]
    for item
    in versioned_files
)

print(
    "Version suffix counts:",
    dict(
        sorted(
            version_counts.items()
        )
    ),
)

for item in versioned_files[
    :120
]:
    print(
        "-",
        item[
            "relative_path"
        ],
        "| v",
        item[
            "version"
        ],
        sep="",
    )

# -------------------------------------------------------------------
# Discover report / acceptance pair conventions
# -------------------------------------------------------------------

report_files = [
    path
    for path in artifact_files
    if (
        "reports/phase2"
        in str(
            path.relative_to(
                repo_root
            )
        )
    )
]

acceptance_files = [
    path
    for path in report_files
    if "acceptance"
    in path.name.lower()
]

quality_report_files = [
    path
    for path in report_files
    if (
        "quality"
        in path.name.lower()
        or "report"
        in path.name.lower()
    )
]

print(
    "\n=== REPORT / ACCEPTANCE CONVENTIONS ==="
)

print(
    "Acceptance files:",
    len(
        acceptance_files
    ),
)

for path in sorted(
    acceptance_files
):
    print(
        "-",
        path.relative_to(
            repo_root
        ),
    )

print(
    "\nReport-like files:",
    len(
        quality_report_files
    ),
)

for path in sorted(
    quality_report_files
):
    print(
        "-",
        path.relative_to(
            repo_root
        ),
    )

# -------------------------------------------------------------------
# Discover processed dataset naming
# -------------------------------------------------------------------

processed_files = []

if processed_dir.exists():
    processed_files = sorted(
        path
        for path in processed_dir.iterdir()
        if path.is_file()
    )

print(
    "\n=== PROCESSED DATASET NAMING ==="
)

for path in processed_files:
    print(
        "-",
        path.name,
        "| bytes:",
        path.stat().st_size,
    )

# -------------------------------------------------------------------
# Discover checkpoint conventions
# -------------------------------------------------------------------

checkpoint_files = []

if checkpoint_directory_exists:
    checkpoint_files = sorted(
        path
        for path
        in checkpoint_dir.iterdir()
        if path.is_file()
    )

print(
    "\n=== CHECKPOINT NAMING ==="
)

print(
    "Checkpoint directory exists:",
    checkpoint_directory_exists,
)

if checkpoint_files:
    for path in checkpoint_files:
        print(
            "-",
            path.name,
        )
else:
    print(
        "- NONE"
    )

if not checkpoint_directory_exists:
    print(
        "Interpretation: checkpoint directory is not "
        "currently materialized in this repository snapshot."
    )

    print(
        "Roadmap mentions checkpoint artifacts, "
        "but no repository-local naming convention "
        "can be inferred from existing files."
    )

# -------------------------------------------------------------------
# Search specifically for any existing NB09 / dedup / quarantine artifact
# -------------------------------------------------------------------

nb09_candidate_terms = [
    "09_",
    "dedup",
    "deduplication",
    "quarantine",
    "canonical_relation",
    "quality_score",
    "quality_scoring",
]

existing_nb09_artifact_candidates = []

for path in artifact_files:
    relative_path = str(
        path.relative_to(
            repo_root
        )
    )

    lower_path = (
        relative_path.lower()
    )

    if any(
        term.lower()
        in lower_path
        for term
        in nb09_candidate_terms
    ):
        existing_nb09_artifact_candidates.append(
            relative_path
        )

print(
    "\n=== EXISTING NB09 / DEDUP / QUARANTINE CANDIDATES ==="
)

if existing_nb09_artifact_candidates:
    for relative_path in sorted(
        existing_nb09_artifact_candidates
    ):
        print(
            "-",
            relative_path,
        )
else:
    print(
        "- NONE"
    )

# -------------------------------------------------------------------
# Derive evidence-backed naming observations only
# -------------------------------------------------------------------

observations = []

processed_v1_jsonl_count = sum(
    (
        path.suffix
        == ".jsonl"
        and "_v1"
        in path.stem
    )
    for path
    in processed_files
)

if processed_v1_jsonl_count:
    observations.append(
        (
            "PROCESSED_DATASET_VERSION_SUFFIX",
            "_v1.jsonl",
            processed_v1_jsonl_count,
        )
    )

acceptance_v1_count = sum(
    "_acceptance_v1.json"
    in path.name
    for path
    in acceptance_files
)

if acceptance_v1_count:
    observations.append(
        (
            "ACCEPTANCE_REPORT_SUFFIX",
            "_acceptance_v1.json",
            acceptance_v1_count,
        )
    )

report_v1_count = sum(
    (
        "_report_v1.json"
        in path.name
        or "_report_v1.md"
        in path.name
    )
    for path
    in report_files
)

if report_v1_count:
    observations.append(
        (
            "REPORT_VERSION_SUFFIX",
            "_report_v1",
            report_v1_count,
        )
    )

if not checkpoint_directory_exists:
    observations.append(
        (
            "CHECKPOINT_CONVENTION_STATUS",
            "NO_EXISTING_REPOSITORY_LOCAL_CONVENTION",
            0,
        )
    )

if not existing_nb09_artifact_candidates:
    observations.append(
        (
            "NB09_EXISTING_ARTIFACT_STATUS",
            "NONE_DISCOVERED",
            0,
        )
    )

print(
    "\n=== EVIDENCE-BACKED NAMING OBSERVATIONS ==="
)

if observations:
    for (
        observation_type,
        pattern,
        evidence_count,
    ) in observations:
        print(
            "-",
            observation_type,
            "| pattern:",
            pattern,
            "| evidence files:",
            evidence_count,
        )
else:
    print(
        "- NONE"
    )

# -------------------------------------------------------------------
# Validation
# -------------------------------------------------------------------

required_discovery_roots_present = all(
    directory.exists()
    for directory
    in required_discovery_directories
)

checkpoint_state_inspected = (
    checkpoint_directory_exists
    or not checkpoint_directory_exists
)

artifact_inventory_nonempty = (
    len(
        artifact_files
    )
    > 0
)

processed_convention_found = (
    len(
        processed_files
    )
    > 0
)

acceptance_convention_found = (
    len(
        acceptance_files
    )
    > 0
)

report_convention_found = (
    len(
        report_files
    )
    > 0
)

version_convention_found = (
    len(
        versioned_files
    )
    > 0
)

nb09_prior_state_inspected = (
    isinstance(
        existing_nb09_artifact_candidates,
        list,
    )
)

print(
    "\nValidation:"
)

print(
    "Required discovery roots present "
    "(processed/interim/reports):",
    "PASS"
    if required_discovery_roots_present
    else "FAIL",
)

print(
    "Checkpoint directory state inspected:",
    "PASS"
    if checkpoint_state_inspected
    else "FAIL",
)

print(
    "Artifact inventory discovered:",
    "PASS"
    if artifact_inventory_nonempty
    else "FAIL",
)

print(
    "Processed dataset naming evidence found:",
    "PASS"
    if processed_convention_found
    else "FAIL",
)

print(
    "Acceptance naming evidence found:",
    "PASS"
    if acceptance_convention_found
    else "FAIL",
)

print(
    "Report naming evidence found:",
    "PASS"
    if report_convention_found
    else "FAIL",
)

print(
    "Version suffix convention found:",
    "PASS"
    if version_convention_found
    else "FAIL",
)

print(
    "Existing NB09 artifact state inspected:",
    "PASS"
    if nb09_prior_state_inspected
    else "FAIL",
)

cell_gate = (
    required_discovery_roots_present
    and checkpoint_state_inspected
    and artifact_inventory_nonempty
    and processed_convention_found
    and acceptance_convention_found
    and report_convention_found
    and version_convention_found
    and nb09_prior_state_inspected
)

print(
    "CODE CELL 35 GATE:",
    "PASS"
    if cell_gate
    else "FAIL",
)
```

<div style="direction: rtl; white-space: normal; line-height: 1.7;">

<h3>Freeze قرارداد Quality، Quarantine، Processing Acceptance و Release Eligibility</h3>

<p>
Discoveryهای Notebook 09 اکنون برای تعریف Policy اجرایی کافی هستند.
این مرحله Policy نهایی را در حافظه تثبیت می‌کند، اما هنوز هیچ Artifact روی Disk نمی‌نویسد.
</p>

<h4>1. Processing Acceptance</h4>

<p>
یک Document برای ادامه مراحل داخلی فاز 2 قابل استفاده است اگر Hard Structural Failure نداشته باشد.
</p>

<p>
Hard Failureهای این Notebook عبارت‌اند از:
</p>

<ul>
<li>Schema invalidity</li>
<li>متن خالی</li>
<li>Locator identity تکراری</li>
<li>Lineage hash inconsistency</li>
</ul>

<p>
Discovery فعلی نشان داده است که Population این گروه صفر است.
</p>

<h4>2. Review Candidate</h4>

<p>
Review Candidate یک Document قابل پردازش است که Evidence غیرعادی یا نیازمند بررسی انسانی دارد.
Review به معنی Quarantine خودکار نیست.
</p>

<p>
در Snapshot فعلی، 7 Segment برنامه «خانه تو» با Labelهای غیرمنتظره
<code>شرکت‌کننده 1</code>
و
<code>شرکت‌کننده 2</code>
در این گروه قرار می‌گیرند.
</p>

<h4>3. Quarantine</h4>

<p>
برای Notebook 09 فقط Hard Failure قطعی می‌تواند Quarantine ساختاری ایجاد کند.
Soft Metadata Gap، Speaker Unresolved، Exact Duplicate، Near-Duplicate یا Review Flag به‌تنهایی Quarantine ایجاد نمی‌کنند.
</p>

<p>
Rights یک محور جداگانه است:
Repository اجازه Processing را می‌دهد، ولی Release رسمی را تا تکمیل Rights Evidence گیت می‌کند.
بنابراین Rights unresolved با Quarantine کیفیتی یکی گرفته نمی‌شود.
</p>

<h4>4. Release Eligibility</h4>

<p>
Release Eligibility از Processing Acceptance جدا است.
یک Document می‌تواند برای Processing پذیرفته شود اما به علت Rights Evidence وارد Release رسمی نشود.
</p>

<p>
بر اساس Policy فعلی:
</p>

<ul>
<li><code>EVIDENCE_REQUIRED</code> → Processing مجاز، Release گیت‌شده.</li>
<li>Website با Rights status نامشخص → Processing مجاز، Release rights review لازم.</li>
<li>وجود Source در Repository به‌تنهایی Rights Evidence نیست.</li>
</ul>

<h4>5. Dedup Policy</h4>

<ul>
<li>Exact Duplicate یک Relation است، نه حذف خودکار.</li>
<li>Near-Duplicate یک Relation است، نه حذف خودکار.</li>
<li>Cross-source similarity هرگز به‌تنهایی Merge یا Delete ایجاد نمی‌کند.</li>
<li>Author، Speaker، Source Identity، Locator، Rights و Lineage حفظ می‌شوند.</li>
</ul>

<h4>6. Authority Policy</h4>

<p>
Authority در Quality Score با Completeness مخلوط نمی‌شود.
یک Segment مجری یا Participant ممکن است Quality کاملی داشته باشد ولی Primary Knowledge Authority نباشد.
</p>

<p>
بنابراین Authority به‌صورت Categorical Evidence مستقل نگهداری می‌شود.
</p>

<h4>7. Quality Scoring</h4>

<p>
Roadmap Quality Scoring را الزام کرده است، اما Repository هیچ Weight عددی برای Source، Authority یا Metadata تعریف نکرده است.
بنابراین Weight دلخواه ساخته نمی‌شود.
</p>

<p>
در نسخه اول، Quality Score صرفاً یک
<strong>Unweighted Structural Completeness Score</strong>
است.
</p>

<p>
پنج Component مستقل و قابل دفاع:
</p>

<ol>
<li>Schema validity</li>
<li>Non-empty text</li>
<li>Locator integrity</li>
<li>Lineage integrity</li>
<li>Text hash integrity</li>
</ol>

<p>
هر Component فقط مقدار 0 یا 1 دارد.
</p>

<p>
فرمول:
</p>

<p>
<code>quality_completeness_score = passed_components / 5</code>
</p>

<p>
این Score:
</p>

<ul>
<li>Authority Score نیست.</li>
<li>Rights Score نیست.</li>
<li>Semantic Quality Score نیست.</li>
<li>Retrieval Score نیست.</li>
<li>Weight دلخواه ندارد.</li>
</ul>

<p>
Soft Metadata Gapها، Rights، Authority، Review و Dedup به‌صورت Evidence جدا در کنار Score نگهداری می‌شوند.
</p>

<h4>معیار PASS</h4>

<p>
تمام 19,456 Document باید بر اساس همین Policy طبقه‌بندی شوند،
Quality Score فقط از پنج Component ساختاری محاسبه شود،
Quarantine فقط از Hard Failure ساخته شود،
و Processing Acceptance با Release Eligibility مخلوط نشود.
</p>

</div>

```python
# Code Cell 36

from collections import Counter, defaultdict
import hashlib
import json

print(
    "=== NOTEBOOK 09 POLICY FREEZE ==="
)

# -------------------------------------------------------------------
# Versioned policy contract
# -------------------------------------------------------------------

nb09_policy_contract = {
    "policy_name": (
        "deduplication_and_quality_policy_v1"
    ),
    "notebook": (
        "09_deduplication_and_quality.ipynb"
    ),
    "processing_acceptance": {
        "rule": (
            "NO_HARD_STRUCTURAL_FAILURE"
        ),
        "hard_failure_reasons": [
            "SCHEMA_INVALID",
            "EMPTY_TEXT",
            "DUPLICATE_FULL_LOCATOR",
            "LINEAGE_HASH_MISMATCH",
            "TEXT_HASH_MISMATCH",
        ],
    },
    "review": {
        "automatic_quarantine": False,
        "known_review_flags": [
            "KHANE_TO_UNEXPECTED_PARTICIPANT_LABEL",
        ],
    },
    "quarantine": {
        "rule": (
            "HARD_STRUCTURAL_FAILURE_ONLY"
        ),
        "soft_metadata_gap_is_quarantine": False,
        "exact_duplicate_is_quarantine": False,
        "near_duplicate_is_quarantine": False,
        "unresolved_speaker_is_quarantine": False,
        "review_candidate_is_quarantine": False,
    },
    "rights": {
        "processing_allowed_when_unresolved": True,
        "release_rights_gated": True,
        "evidence_required_is_quality_failure": False,
        "repository_presence_is_rights_evidence": False,
    },
    "dedup": {
        "exact_duplicate_action": (
            "PRESERVE_AND_RELATE"
        ),
        "near_duplicate_action": (
            "PRESERVE_AND_RELATE"
        ),
        "cross_source_action": (
            "PRESERVE_SOURCE_IDENTITY_AND_LINEAGE"
        ),
        "automatic_merge": False,
        "automatic_delete": False,
    },
    "authority": {
        "representation": (
            "CATEGORICAL_EVIDENCE"
        ),
        "included_in_quality_score": False,
        "primary_knowledge_authority": (
            "Dr. Komeil Roudi"
        ),
    },
    "quality_score": {
        "name": (
            "structural_completeness_score_v1"
        ),
        "aggregation": (
            "UNWEIGHTED_BINARY_COMPONENT_MEAN"
        ),
        "components": [
            "SCHEMA_VALID",
            "TEXT_NONEMPTY",
            "LOCATOR_INTEGRITY",
            "LINEAGE_INTEGRITY",
            "TEXT_HASH_INTEGRITY",
        ],
        "component_weighting": (
            "NONE_EQUAL_BINARY_COMPONENTS"
        ),
        "authority_included": False,
        "rights_included": False,
        "soft_metadata_included": False,
        "dedup_included": False,
        "review_flags_included": False,
    },
}

print(
    json.dumps(
        nb09_policy_contract,
        ensure_ascii=False,
        indent=2,
        sort_keys=True,
    )
)

# -------------------------------------------------------------------
# Recompute text hash integrity independently
# -------------------------------------------------------------------

text_hash_integrity_lookup = {}

text_hash_mismatch_ids = []

for record in quality_records:
    document_id = record[
        "document_id"
    ]

    text = record[
        "text"
    ]

    recomputed_hash = (
        hashlib.sha256(
            text.encode(
                "utf-8"
            )
        ).hexdigest()
    )

    stored_hash = record[
        "text_sha256"
    ]

    valid = (
        recomputed_hash
        == stored_hash
    )

    text_hash_integrity_lookup[
        document_id
    ] = valid

    if not valid:
        text_hash_mismatch_ids.append(
            document_id
        )

print(
    "\n=== TEXT HASH INTEGRITY REVALIDATION ==="
)

print(
    "Text hash mismatches:",
    len(
        text_hash_mismatch_ids
    ),
)

# -------------------------------------------------------------------
# Reuse schema result from Cell 30
# All documents were independently schema-valid.
# -------------------------------------------------------------------

schema_valid_lookup = {
    record[
        "document_id"
    ]: True
    for record in quality_records
}

# -------------------------------------------------------------------
# Build final policy rows
# -------------------------------------------------------------------

quality_policy_rows = []

for record, evidence_row in zip(
    quality_records,
    quality_evidence_rows,
):
    document_id = record[
        "document_id"
    ]

    source_type = record[
        "source_type"
    ]

    # ---------------------------------------------------------------
    # Five binary structural completeness components
    # ---------------------------------------------------------------

    schema_valid = (
        schema_valid_lookup.get(
            document_id,
            False,
        )
    )

    text_nonempty = (
        isinstance(
            record[
                "text"
            ],
            str,
        )
        and record[
            "text"
        ].strip()
        != ""
    )

    locator_integrity = (
        "DUPLICATE_FULL_LOCATOR"
        not in evidence_row[
            "hard_failure_reasons"
        ]
    )

    lineage_integrity = (
        "LINEAGE_HASH_MISMATCH"
        not in evidence_row[
            "hard_failure_reasons"
        ]
    )

    text_hash_integrity = (
        text_hash_integrity_lookup[
            document_id
        ]
    )

    component_values = {
        "SCHEMA_VALID": (
            int(
                schema_valid
            )
        ),
        "TEXT_NONEMPTY": (
            int(
                text_nonempty
            )
        ),
        "LOCATOR_INTEGRITY": (
            int(
                locator_integrity
            )
        ),
        "LINEAGE_INTEGRITY": (
            int(
                lineage_integrity
            )
        ),
        "TEXT_HASH_INTEGRITY": (
            int(
                text_hash_integrity
            )
        ),
    }

    passed_components = sum(
        component_values.values()
    )

    total_components = len(
        component_values
    )

    quality_completeness_score = (
        passed_components
        / total_components
    )

    # ---------------------------------------------------------------
    # Hard failure recomputation
    # ---------------------------------------------------------------

    hard_failure_reasons = list(
        evidence_row[
            "hard_failure_reasons"
        ]
    )

    if (
        not text_hash_integrity
        and "TEXT_HASH_MISMATCH"
        not in hard_failure_reasons
    ):
        hard_failure_reasons.append(
            "TEXT_HASH_MISMATCH"
        )

    hard_failure = bool(
        hard_failure_reasons
    )

    # ---------------------------------------------------------------
    # Processing acceptance
    # ---------------------------------------------------------------

    if hard_failure:
        processing_decision = (
            "QUARANTINED_HARD_FAILURE"
        )
    else:
        processing_decision = (
            "ACCEPTED_FOR_PROCESSING"
        )

    # ---------------------------------------------------------------
    # Rights / release eligibility
    # ---------------------------------------------------------------

    rights_class = evidence_row[
        "rights_evidence_class"
    ]

    if rights_class in {
        "RIGHTS_EVIDENCE_REQUIRED",
        "RIGHTS_STATUS_NOT_ESTABLISHED",
    }:
        release_eligibility = (
            "RELEASE_BLOCKED_PENDING_RIGHTS"
        )
    else:
        release_eligibility = (
            "RELEASE_RIGHTS_GATE_SATISFIED"
        )

    # ---------------------------------------------------------------
    # Review
    # ---------------------------------------------------------------

    review_required = bool(
        evidence_row[
            "review_flags"
        ]
    )

    # ---------------------------------------------------------------
    # Final in-memory policy record
    # ---------------------------------------------------------------

    quality_policy_rows.append(
        {
            "document_id": (
                document_id
            ),
            "source_type": (
                source_type
            ),
            "source_id": (
                record[
                    "source_id"
                ]
            ),
            "quality_score_name": (
                "structural_completeness_score_v1"
            ),
            "quality_components": (
                component_values
            ),
            "quality_components_passed": (
                passed_components
            ),
            "quality_components_total": (
                total_components
            ),
            "quality_completeness_score": (
                quality_completeness_score
            ),
            "hard_failure": (
                hard_failure
            ),
            "hard_failure_reasons": (
                hard_failure_reasons
            ),
            "processing_decision": (
                processing_decision
            ),
            "release_eligibility": (
                release_eligibility
            ),
            "rights_evidence_class": (
                rights_class
            ),
            "soft_metadata_flags": list(
                evidence_row[
                    "soft_metadata_flags"
                ]
            ),
            "authority_class": (
                evidence_row[
                    "authority_class"
                ]
            ),
            "authority_basis": (
                evidence_row[
                    "authority_basis"
                ]
            ),
            "review_required": (
                review_required
            ),
            "review_flags": list(
                evidence_row[
                    "review_flags"
                ]
            ),
            "dedup_relation_class": (
                evidence_row[
                    "dedup_relation_class"
                ]
            ),
        }
    )

# -------------------------------------------------------------------
# Global profiles
# -------------------------------------------------------------------

print(
    "\n=== QUALITY SCORE PROFILE ==="
)

score_counts = Counter(
    row[
        "quality_completeness_score"
    ]
    for row
    in quality_policy_rows
)

for score, count in sorted(
    score_counts.items(),
    reverse=True,
):
    print(
        "-",
        score,
        ":",
        count,
    )

component_failure_counts = Counter()

for row in quality_policy_rows:
    for component, value in (
        row[
            "quality_components"
        ].items()
    ):
        if value == 0:
            component_failure_counts[
                component
            ] += 1

print(
    "\nComponent failures:"
)

if component_failure_counts:
    for component, count in (
        component_failure_counts.most_common()
    ):
        print(
            "-",
            component,
            ":",
            count,
        )
else:
    print(
        "- NONE"
    )

# -------------------------------------------------------------------
# Processing profile
# -------------------------------------------------------------------

processing_decision_counts = Counter(
    row[
        "processing_decision"
    ]
    for row
    in quality_policy_rows
)

print(
    "\n=== PROCESSING DECISION PROFILE ==="
)

for value, count in (
    processing_decision_counts.most_common()
):
    print(
        "-",
        value,
        ":",
        count,
    )

# -------------------------------------------------------------------
# Release profile
# -------------------------------------------------------------------

release_eligibility_counts = Counter(
    row[
        "release_eligibility"
    ]
    for row
    in quality_policy_rows
)

print(
    "\n=== RELEASE ELIGIBILITY PROFILE ==="
)

for value, count in (
    release_eligibility_counts.most_common()
):
    print(
        "-",
        value,
        ":",
        count,
    )

# -------------------------------------------------------------------
# Review profile
# -------------------------------------------------------------------

review_required_count = sum(
    row[
        "review_required"
    ]
    for row
    in quality_policy_rows
)

print(
    "\n=== REVIEW PROFILE ==="
)

print(
    "Review-required documents:",
    review_required_count,
)

# -------------------------------------------------------------------
# Source-level profile
# -------------------------------------------------------------------

print(
    "\n=== SOURCE-LEVEL POLICY PROFILE ==="
)

for source_type in [
    "website",
    "book",
    "transcript",
]:
    rows = [
        row
        for row in quality_policy_rows
        if row[
            "source_type"
        ]
        == source_type
    ]

    print(
        "\nSOURCE:",
        source_type,
    )

    print(
        "Documents:",
        len(
            rows
        ),
    )

    print(
        "Processing decisions:",
        dict(
            Counter(
                row[
                    "processing_decision"
                ]
                for row in rows
            )
        ),
    )

    print(
        "Release eligibility:",
        dict(
            Counter(
                row[
                    "release_eligibility"
                ]
                for row in rows
            )
        ),
    )

    print(
        "Quality scores:",
        dict(
            Counter(
                row[
                    "quality_completeness_score"
                ]
                for row in rows
            )
        ),
    )

    print(
        "Review required:",
        sum(
            row[
                "review_required"
            ]
            for row in rows
        ),
    )

# -------------------------------------------------------------------
# Explicit quarantine and accepted populations
# -------------------------------------------------------------------

processing_accepted_rows = [
    row
    for row in quality_policy_rows
    if row[
        "processing_decision"
    ]
    == "ACCEPTED_FOR_PROCESSING"
]

hard_quarantine_rows = [
    row
    for row in quality_policy_rows
    if row[
        "processing_decision"
    ]
    == "QUARANTINED_HARD_FAILURE"
]

release_blocked_rows = [
    row
    for row in quality_policy_rows
    if row[
        "release_eligibility"
    ]
    == "RELEASE_BLOCKED_PENDING_RIGHTS"
]

release_rights_satisfied_rows = [
    row
    for row in quality_policy_rows
    if row[
        "release_eligibility"
    ]
    == "RELEASE_RIGHTS_GATE_SATISFIED"
]

print(
    "\n=== POLICY POPULATIONS ==="
)

print(
    "Accepted for processing:",
    len(
        processing_accepted_rows
    ),
)

print(
    "Hard quarantine:",
    len(
        hard_quarantine_rows
    ),
)

print(
    "Release blocked pending rights:",
    len(
        release_blocked_rows
    ),
)

print(
    "Release rights gate satisfied:",
    len(
        release_rights_satisfied_rows
    ),
)

# -------------------------------------------------------------------
# Validation
# -------------------------------------------------------------------

population_accounted = (
    len(
        quality_policy_rows
    )
    == 19456
)

all_scores_bounded = all(
    0.0
    <= row[
        "quality_completeness_score"
    ]
    <= 1.0
    for row in quality_policy_rows
)

all_scores_component_derived = all(
    row[
        "quality_completeness_score"
    ]
    == (
        row[
            "quality_components_passed"
        ]
        / row[
            "quality_components_total"
        ]
    )
    for row in quality_policy_rows
)

hard_quarantine_matches_failures = all(
    (
        row[
            "processing_decision"
        ]
        == "QUARANTINED_HARD_FAILURE"
    )
    == row[
        "hard_failure"
    ]
    for row in quality_policy_rows
)

processing_population_partitioned = (
    len(
        processing_accepted_rows
    )
    + len(
        hard_quarantine_rows
    )
    == 19456
)

release_population_partitioned = (
    len(
        release_blocked_rows
    )
    + len(
        release_rights_satisfied_rows
    )
    == 19456
)

review_population_preserved = (
    review_required_count
    == 7
)

text_hash_integrity_clean = (
    len(
        text_hash_mismatch_ids
    )
    == 0
)

no_dataset_mutation = (
    len(
        quality_records
    )
    == 19456
)

print(
    "\nValidation:"
)

print(
    "All documents classified:",
    "PASS"
    if population_accounted
    else "FAIL",
)

print(
    "All quality scores bounded [0,1]:",
    "PASS"
    if all_scores_bounded
    else "FAIL",
)

print(
    "All quality scores derived only from 5 binary components:",
    "PASS"
    if all_scores_component_derived
    else "FAIL",
)

print(
    "Hard quarantine exactly matches hard failures:",
    "PASS"
    if hard_quarantine_matches_failures
    else "FAIL",
)

print(
    "Processing population partitioned:",
    "PASS"
    if processing_population_partitioned
    else "FAIL",
)

print(
    "Release population partitioned:",
    "PASS"
    if release_population_partitioned
    else "FAIL",
)

print(
    "7 review candidates preserved without quarantine:",
    "PASS"
    if review_population_preserved
    else "FAIL",
)

print(
    "Independent text hash integrity remains clean:",
    "PASS"
    if text_hash_integrity_clean
    else "FAIL",
)

print(
    "Input dataset preserved without mutation:",
    "PASS"
    if no_dataset_mutation
    else "FAIL",
)

cell_gate = (
    population_accounted
    and all_scores_bounded
    and all_scores_component_derived
    and hard_quarantine_matches_failures
    and processing_population_partitioned
    and release_population_partitioned
    and review_population_preserved
    and text_hash_integrity_clean
    and no_dataset_mutation
)

print(
    "CODE CELL 36 GATE:",
    "PASS"
    if cell_gate
    else "FAIL",
)
```

Near-Duplicate Candidateها بدون Promotion خودکار حفظ شوند

```python
# Code Cell 37

from collections import Counter, defaultdict
import hashlib

print("=== FINAL IN-MEMORY OUTPUT PROJECTIONS ===")

# -------------------------------------------------------------------
# 1. Processing / quarantine / review / release populations
# -------------------------------------------------------------------

accepted_processing_ids = {
    row["document_id"]
    for row in quality_policy_rows
    if row["processing_decision"] == "ACCEPTED_FOR_PROCESSING"
}

hard_quarantine_ids = {
    row["document_id"]
    for row in quality_policy_rows
    if row["processing_decision"] == "QUARANTINED_HARD_FAILURE"
}

review_queue_ids = {
    row["document_id"]
    for row in quality_policy_rows
    if row["review_required"]
}

release_blocked_ids = {
    row["document_id"]
    for row in quality_policy_rows
    if row["release_eligibility"] == "RELEASE_BLOCKED_PENDING_RIGHTS"
}

release_rights_satisfied_ids = {
    row["document_id"]
    for row in quality_policy_rows
    if row["release_eligibility"] == "RELEASE_RIGHTS_GATE_SATISFIED"
}

print("Accepted for processing:", len(accepted_processing_ids))
print("Hard quarantine:", len(hard_quarantine_ids))
print("Review queue:", len(review_queue_ids))
print("Release blocked pending rights:", len(release_blocked_ids))
print("Release rights gate satisfied:", len(release_rights_satisfied_ids))

accepted_processing_records = [
    record
    for record in quality_records
    if record["document_id"] in accepted_processing_ids
]

hard_quarantine_records = [
    record
    for record in quality_records
    if record["document_id"] in hard_quarantine_ids
]

review_queue_records = [
    row
    for row in quality_policy_rows
    if row["document_id"] in review_queue_ids
]

# -------------------------------------------------------------------
# 2. Exact duplicate relation ledger
# -------------------------------------------------------------------

records_by_text_hash = defaultdict(list)

for record in quality_records:
    records_by_text_hash[
        record["text_sha256"]
    ].append(record)

exact_duplicate_groups_final = []

for text_hash in sorted(records_by_text_hash):
    members = records_by_text_hash[text_hash]

    if len(members) <= 1:
        continue

    member_document_ids = sorted(
        record["document_id"]
        for record in members
    )

    source_types = sorted(
        {
            record["source_type"]
            for record in members
        }
    )

    source_ids = sorted(
        {
            record["source_id"]
            for record in members
        }
    )

    if len(source_types) > 1:
        relation_scope = "CROSS_SOURCE_TYPE"

    elif len(source_ids) > 1:
        relation_scope = (
            "SAME_SOURCE_TYPE_DIFFERENT_SOURCE_ID"
        )

    else:
        relation_scope = "SAME_SOURCE_ID"

    relation_group_id = (
        "exact:"
        + hashlib.sha256(
            (
                text_hash
                + "|"
                + "|".join(member_document_ids)
            ).encode("utf-8")
        ).hexdigest()[:24]
    )

    exact_duplicate_groups_final.append(
        {
            "relation_group_id": relation_group_id,
            "relation_type": "EXACT_TEXT_DUPLICATE_RELATION",
            "text_sha256": text_hash,
            "member_count": len(members),
            "member_document_ids": member_document_ids,
            "source_types": source_types,
            "source_ids": source_ids,
            "relation_scope": relation_scope,
            "automatic_merge": False,
            "automatic_delete": False,
            "canonical_document_id": None,
            "canonical_status": (
                "NOT_SELECTED_WITHOUT_INDEPENDENT_EVIDENCE"
            ),
        }
    )

exact_group_count = len(
    exact_duplicate_groups_final
)

exact_member_count = sum(
    group["member_count"]
    for group in exact_duplicate_groups_final
)

exact_scope_counts = Counter(
    group["relation_scope"]
    for group in exact_duplicate_groups_final
)

print("\n=== EXACT RELATION LEDGER ===")
print("Exact relation groups:", exact_group_count)
print("Exact relation members:", exact_member_count)
print("Relation scopes:", dict(exact_scope_counts))

# -------------------------------------------------------------------
# 3. Use the collections actually discovered in the previous run
# -------------------------------------------------------------------

within_collection_exists = (
    "pair_feature_records" in globals()
    and isinstance(pair_feature_records, list)
)

cross_collection_exists = (
    "cross_pair_results" in globals()
    and isinstance(cross_pair_results, list)
)

print("\n=== PAIR COLLECTION CONTRACT ===")

print(
    "pair_feature_records exists:",
    within_collection_exists,
)

print(
    "cross_pair_results exists:",
    cross_collection_exists,
)

if within_collection_exists:
    print(
        "pair_feature_records rows:",
        len(pair_feature_records),
    )

if cross_collection_exists:
    print(
        "cross_pair_results rows:",
        len(cross_pair_results),
    )

# -------------------------------------------------------------------
# 4. Independently validate the within-source collection
# -------------------------------------------------------------------

EXPECTED_WITHIN_BANDS = Counter(
    {
        "GE_0_98": 52,
        "0_95_TO_0_98": 46,
        "0_90_TO_0_95": 96,
        "0_80_TO_0_90": 461,
        "LT_0_80": 15274,
    }
)

def similarity_band(score):
    if score >= 0.98:
        return "GE_0_98"

    if score >= 0.95:
        return "0_95_TO_0_98"

    if score >= 0.90:
        return "0_90_TO_0_95"

    if score >= 0.80:
        return "0_80_TO_0_90"

    return "LT_0_80"


within_required_keys = {
    "document_id",
    "candidate_document_id",
    "source_type",
    "similarity",
    "sequence_matcher",
    "token_jaccard",
    "length_ratio",
}

within_schema_valid = (
    within_collection_exists
    and len(pair_feature_records) == 15929
    and all(
        within_required_keys.issubset(
            row.keys()
        )
        for row in pair_feature_records
    )
)

if within_schema_valid:
    within_empirical_bands = Counter(
        similarity_band(
            float(row["similarity"])
        )
        for row in pair_feature_records
    )
else:
    within_empirical_bands = Counter()

within_empirical_profile_valid = (
    within_empirical_bands
    == EXPECTED_WITHIN_BANDS
)

print(
    "\nWithin-source required fields valid:",
    within_schema_valid,
)

print(
    "Within-source empirical bands:",
    dict(within_empirical_bands),
)

print(
    "Within-source empirical profile valid:",
    within_empirical_profile_valid,
)

# -------------------------------------------------------------------
# 5. Independently validate cross-source collection
# -------------------------------------------------------------------

EXPECTED_CROSS_PAIRS = Counter(
    {
        ("book", "transcript"): 18146,
        ("book", "website"): 3096,
        ("transcript", "website"): 16554,
    }
)

cross_required_keys = {
    "document_id",
    "candidate_document_id",
    "query_source_type",
    "target_source_type",
    "similarity",
    "sequence_matcher",
    "token_jaccard",
    "length_ratio",
}

cross_schema_valid = (
    cross_collection_exists
    and len(cross_pair_results) == 37796
    and all(
        cross_required_keys.issubset(
            row.keys()
        )
        for row in cross_pair_results
    )
)

if cross_schema_valid:
    cross_empirical_pairs = Counter(
        tuple(
            sorted(
                (
                    row["query_source_type"],
                    row["target_source_type"],
                )
            )
        )
        for row in cross_pair_results
    )
else:
    cross_empirical_pairs = Counter()

cross_empirical_profile_valid = (
    cross_empirical_pairs
    == EXPECTED_CROSS_PAIRS
)

print(
    "\nCross-source required fields valid:",
    cross_schema_valid,
)

print(
    "Cross-source empirical source pairs:",
    {
        "|".join(pair): count
        for pair, count
        in sorted(
            cross_empirical_pairs.items()
        )
    },
)

print(
    "Cross-source empirical profile valid:",
    cross_empirical_profile_valid,
)

# -------------------------------------------------------------------
# 6. Document lookup
# -------------------------------------------------------------------

document_lookup_for_relations = {
    record["document_id"]: record
    for record in quality_records
}

# -------------------------------------------------------------------
# 7. Near-duplicate candidate ledger
# -------------------------------------------------------------------

near_duplicate_candidate_relations = []
near_pair_failures = []

if within_schema_valid:
    for pair_index, pair in enumerate(
        pair_feature_records
    ):
        try:
            document_id_a = pair["document_id"]

            document_id_b = (
                pair["candidate_document_id"]
            )

            similarity = float(
                pair["similarity"]
            )

            source_type = pair["source_type"]

        except (
            KeyError,
            TypeError,
            ValueError,
        ):
            near_pair_failures.append(
                pair_index
            )
            continue

        if (
            document_id_a
            not in document_lookup_for_relations
            or document_id_b
            not in document_lookup_for_relations
        ):
            near_pair_failures.append(
                pair_index
            )
            continue

        source_type_a = (
            document_lookup_for_relations[
                document_id_a
            ]["source_type"]
        )

        source_type_b = (
            document_lookup_for_relations[
                document_id_b
            ]["source_type"]
        )

        if (
            source_type_a != source_type
            or source_type_b != source_type
        ):
            near_pair_failures.append(
                pair_index
            )
            continue

        ordered_ids = sorted(
            (
                document_id_a,
                document_id_b,
            )
        )

        relation_id = (
            "near:"
            + hashlib.sha256(
                (
                    ordered_ids[0]
                    + "|"
                    + ordered_ids[1]
                ).encode("utf-8")
            ).hexdigest()[:24]
        )

        near_duplicate_candidate_relations.append(
            {
                "relation_id": relation_id,
                "relation_type": (
                    "CANDIDATE_NEAR_DUPLICATE_RELATION"
                ),
                "document_id_a": document_id_a,
                "document_id_b": document_id_b,
                "source_type": source_type,
                "source_type_a": source_type_a,
                "source_type_b": source_type_b,
                "char_tfidf_cosine": similarity,
                "sequence_matcher": pair[
                    "sequence_matcher"
                ],
                "token_jaccard": pair[
                    "token_jaccard"
                ],
                "length_ratio": pair[
                    "length_ratio"
                ],
                "production_duplicate_decision": (
                    "NOT_SELECTED"
                ),
                "automatic_merge": False,
                "automatic_delete": False,
                "canonical_document_id": None,
            }
        )

near_similarity_bands = Counter(
    similarity_band(
        relation[
            "char_tfidf_cosine"
        ]
    )
    for relation
    in near_duplicate_candidate_relations
)

print(
    "\n=== NEAR-DUPLICATE CANDIDATE LEDGER ==="
)

print(
    "Candidate relations:",
    len(
        near_duplicate_candidate_relations
    ),
)

print(
    "Pair failures:",
    len(
        near_pair_failures
    ),
)

print(
    "Similarity bands:",
    dict(
        near_similarity_bands
    ),
)

# -------------------------------------------------------------------
# 8. Cross-source candidate ledger
# -------------------------------------------------------------------

cross_source_candidate_relations = []
cross_pair_failures = []

if cross_schema_valid:
    for pair_index, pair in enumerate(
        cross_pair_results
    ):
        try:
            document_id_a = (
                pair["document_id"]
            )

            document_id_b = (
                pair[
                    "candidate_document_id"
                ]
            )

            source_type_a = (
                pair["query_source_type"]
            )

            source_type_b = (
                pair["target_source_type"]
            )

            similarity = float(
                pair["similarity"]
            )

        except (
            KeyError,
            TypeError,
            ValueError,
        ):
            cross_pair_failures.append(
                pair_index
            )
            continue

        if (
            document_id_a
            not in document_lookup_for_relations
            or document_id_b
            not in document_lookup_for_relations
        ):
            cross_pair_failures.append(
                pair_index
            )
            continue

        actual_source_type_a = (
            document_lookup_for_relations[
                document_id_a
            ]["source_type"]
        )

        actual_source_type_b = (
            document_lookup_for_relations[
                document_id_b
            ]["source_type"]
        )

        if (
            actual_source_type_a
            != source_type_a
            or actual_source_type_b
            != source_type_b
        ):
            cross_pair_failures.append(
                pair_index
            )
            continue

        ordered_ids = sorted(
            (
                document_id_a,
                document_id_b,
            )
        )

        relation_id = (
            "cross:"
            + hashlib.sha256(
                (
                    ordered_ids[0]
                    + "|"
                    + ordered_ids[1]
                ).encode("utf-8")
            ).hexdigest()[:24]
        )

        cross_source_candidate_relations.append(
            {
                "relation_id": relation_id,
                "relation_type": (
                    "CROSS_SOURCE_SIMILARITY_CANDIDATE"
                ),
                "document_id_a": document_id_a,
                "document_id_b": document_id_b,
                "source_type_a": source_type_a,
                "source_type_b": source_type_b,
                "char_tfidf_cosine": similarity,
                "sequence_matcher": pair[
                    "sequence_matcher"
                ],
                "token_jaccard": pair[
                    "token_jaccard"
                ],
                "length_ratio": pair[
                    "length_ratio"
                ],
                "interpretation": (
                    "SIMILARITY_EVIDENCE_NOT_DUPLICATE_PROOF"
                ),
                "automatic_merge": False,
                "automatic_delete": False,
                "canonical_document_id": None,
            }
        )

cross_source_pair_type_counts = Counter(
    tuple(
        sorted(
            (
                relation[
                    "source_type_a"
                ],
                relation[
                    "source_type_b"
                ],
            )
        )
    )
    for relation
    in cross_source_candidate_relations
)

print(
    "\n=== CROSS-SOURCE SIMILARITY LEDGER ==="
)

print(
    "Cross-source candidate relations:",
    len(
        cross_source_candidate_relations
    ),
)

print(
    "Pair failures:",
    len(
        cross_pair_failures
    ),
)

print(
    "Source-type pair distribution:",
    {
        "|".join(pair): count
        for pair, count
        in sorted(
            cross_source_pair_type_counts.items()
        )
    },
)

# -------------------------------------------------------------------
# 9. Canonical relation policy
# -------------------------------------------------------------------

no_exact_canonical = all(
    group["canonical_document_id"]
    is None
    for group
    in exact_duplicate_groups_final
)

no_near_canonical = all(
    relation["canonical_document_id"]
    is None
    for relation
    in near_duplicate_candidate_relations
)

no_cross_canonical = all(
    relation["canonical_document_id"]
    is None
    for relation
    in cross_source_candidate_relations
)

print(
    "\n=== CANONICAL RELATION POLICY ==="
)

print(
    "Exact groups with automatic canonical:",
    sum(
        group["canonical_document_id"]
        is not None
        for group
        in exact_duplicate_groups_final
    ),
)

print(
    "Near candidates with automatic canonical:",
    sum(
        relation["canonical_document_id"]
        is not None
        for relation
        in near_duplicate_candidate_relations
    ),
)

print(
    "Cross-source candidates with automatic canonical:",
    sum(
        relation["canonical_document_id"]
        is not None
        for relation
        in cross_source_candidate_relations
    ),
)

# -------------------------------------------------------------------
# 10. Validation
# -------------------------------------------------------------------

accepted_quarantine_disjoint = (
    accepted_processing_ids.isdisjoint(
        hard_quarantine_ids
    )
)

processing_partition_complete = (
    len(accepted_processing_ids)
    + len(hard_quarantine_ids)
    == 19456
)

review_subset_of_processing = (
    review_queue_ids.issubset(
        accepted_processing_ids
    )
)

release_partition_complete = (
    len(release_blocked_ids)
    + len(release_rights_satisfied_ids)
    == 19456
)

release_partitions_disjoint = (
    release_blocked_ids.isdisjoint(
        release_rights_satisfied_ids
    )
)

exact_groups_reproduced = (
    exact_group_count == 192
)

exact_members_reproduced = (
    exact_member_count == 931
)

exact_scope_reproduced = (
    exact_scope_counts
    == Counter(
        {
            "SAME_SOURCE_ID": 91,
            "SAME_SOURCE_TYPE_DIFFERENT_SOURCE_ID": 92,
            "CROSS_SOURCE_TYPE": 9,
        }
    )
)

near_population_reproduced = (
    len(
        near_duplicate_candidate_relations
    )
    == 15929
)

near_bands_reproduced = (
    near_similarity_bands
    == EXPECTED_WITHIN_BANDS
)

cross_population_reproduced = (
    len(
        cross_source_candidate_relations
    )
    == 37796
)

cross_distribution_reproduced = (
    cross_source_pair_type_counts
    == EXPECTED_CROSS_PAIRS
)

near_failures_zero = (
    len(near_pair_failures) == 0
)

cross_failures_zero = (
    len(cross_pair_failures) == 0
)

no_automatic_canonicalization = (
    no_exact_canonical
    and no_near_canonical
    and no_cross_canonical
)

no_input_mutation = (
    len(quality_records) == 19456
    and (
        len(accepted_processing_records)
        + len(hard_quarantine_records)
        == 19456
    )
)

print("\nValidation:")

checks = {
    "Accepted and hard-quarantine populations disjoint":
        accepted_quarantine_disjoint,

    "Processing population complete":
        processing_partition_complete,

    "Review queue remains subset of accepted processing population":
        review_subset_of_processing,

    "Release eligibility population complete":
        release_partition_complete,

    "Release partitions disjoint":
        release_partitions_disjoint,

    "Exact duplicate groups reproduced (192)":
        exact_groups_reproduced,

    "Exact duplicate members reproduced (931)":
        exact_members_reproduced,

    "Exact duplicate relation scopes reproduced":
        exact_scope_reproduced,

    "pair_feature_records contract validated":
        (
            within_schema_valid
            and within_empirical_profile_valid
        ),

    "cross_pair_results contract validated":
        (
            cross_schema_valid
            and cross_empirical_profile_valid
        ),

    "Within-source pair failures zero":
        near_failures_zero,

    "Near-duplicate candidate population reproduced (15,929)":
        near_population_reproduced,

    "Near similarity bands reproduced":
        near_bands_reproduced,

    "Cross-source pair failures zero":
        cross_failures_zero,

    "Cross-source candidate population reproduced (37,796)":
        cross_population_reproduced,

    "Cross-source source-pair distribution reproduced":
        cross_distribution_reproduced,

    "No automatic canonical document selected":
        no_automatic_canonicalization,

    "Input dataset preserved without mutation":
        no_input_mutation,
}

for name, passed in checks.items():
    print(
        name + ":",
        "PASS" if passed else "FAIL",
    )

cell_gate = all(
    checks.values()
)

print(
    "CODE CELL 37 GATE:",
    "PASS" if cell_gate else "FAIL",
)
```

```python
# Code Cell 38

from pathlib import Path
from collections import Counter
import json

print(
    "=== NOTEBOOK 09 ARTIFACT CONTRACT FREEZE ==="
)

# -------------------------------------------------------------------
# Output roots discovered and validated previously
# -------------------------------------------------------------------

processed_output_root = (
    repo_root
    / "data/phase2/processed"
)

reports_output_root = (
    repo_root
    / "reports/phase2"
)

# -------------------------------------------------------------------
# Versioned and rerun-safe NB09 artifact contract
# -------------------------------------------------------------------

nb09_artifact_contract = {
    "contract_name": (
        "deduplication_and_quality_artifacts_v1"
    ),
    "notebook": (
        "09_deduplication_and_quality.ipynb"
    ),
    "input": {
        "normalized_dataset": (
            "data/phase2/processed/"
            "documents_normalized_v1.jsonl"
        ),
    },
    "outputs": {
        "accepted_processing_dataset": (
            "data/phase2/processed/"
            "documents_dedup_quality_accepted_v1.jsonl"
        ),
        "hard_quarantine_dataset": (
            "data/phase2/processed/"
            "documents_dedup_quality_quarantine_v1.jsonl"
        ),
        "quality_evidence_records": (
            "data/phase2/processed/"
            "dedup_quality_evidence_records_v1.jsonl"
        ),
        "review_queue": (
            "data/phase2/processed/"
            "dedup_quality_review_queue_v1.jsonl"
        ),
        "exact_duplicate_relations": (
            "data/phase2/processed/"
            "dedup_exact_relations_v1.jsonl"
        ),
        "near_duplicate_candidates": (
            "data/phase2/processed/"
            "dedup_near_candidates_v1.jsonl"
        ),
        "cross_source_similarity_candidates": (
            "data/phase2/processed/"
            "dedup_cross_source_candidates_v1.jsonl"
        ),
        "dedup_quality_report": (
            "reports/phase2/"
            "deduplication_quality_report_v1.json"
        ),
        "dedup_quality_acceptance": (
            "reports/phase2/"
            "deduplication_quality_acceptance_v1.json"
        ),
    },
    "rerun_policy": {
        "mode": (
            "IDEMPOTENT_REBUILD"
        ),
        "planned_existing_outputs_allowed": True,
        "partial_existing_output_set_allowed": True,
        "publication_rule": (
            "REBUILD_IN_STAGING_VALIDATE_THEN_REPLACE"
        ),
    },
    "checkpoint": {
        "status": (
            "DEFERRED_NO_REPOSITORY_LOCAL_CONVENTION"
        ),
        "reason": (
            "artifacts/phase2/checkpoints "
            "was absent during discovery"
        ),
    },
}

print(
    json.dumps(
        nb09_artifact_contract,
        ensure_ascii=False,
        indent=2,
        sort_keys=True,
    )
)

# -------------------------------------------------------------------
# Resolve output paths
# -------------------------------------------------------------------

resolved_output_paths = {
    artifact_name: (
        repo_root
        / relative_path
    )
    for artifact_name, relative_path
    in nb09_artifact_contract[
        "outputs"
    ].items()
}

print(
    "\n=== RESOLVED OUTPUT PATHS ==="
)

existing_output_files = []
invalid_existing_output_objects = []

for artifact_name, path in (
    resolved_output_paths.items()
):
    exists = path.exists()

    print(
        "-",
        artifact_name,
    )

    print(
        " ",
        path.relative_to(
            repo_root
        ),
    )

    print(
        "  exists:",
        exists,
    )

    if exists:
        if path.is_file():
            existing_output_files.append(
                str(
                    path.relative_to(
                        repo_root
                    )
                )
            )
        else:
            invalid_existing_output_objects.append(
                str(
                    path.relative_to(
                        repo_root
                    )
                )
            )

# -------------------------------------------------------------------
# Allowed-root validation
# -------------------------------------------------------------------

def is_relative_to_compat(
    path,
    root,
):
    try:
        path.resolve().relative_to(
            root.resolve()
        )

        return True

    except ValueError:
        return False


path_root_checks = {}

for artifact_name, path in (
    resolved_output_paths.items()
):
    path_root_checks[
        artifact_name
    ] = (
        is_relative_to_compat(
            path,
            processed_output_root,
        )
        or is_relative_to_compat(
            path,
            reports_output_root,
        )
    )

# -------------------------------------------------------------------
# Naming convention checks
# -------------------------------------------------------------------

jsonl_artifacts = {
    "accepted_processing_dataset",
    "hard_quarantine_dataset",
    "quality_evidence_records",
    "review_queue",
    "exact_duplicate_relations",
    "near_duplicate_candidates",
    "cross_source_similarity_candidates",
}

json_report_artifacts = {
    "dedup_quality_report",
    "dedup_quality_acceptance",
}

naming_checks = {}

for artifact_name, path in (
    resolved_output_paths.items()
):
    if artifact_name in jsonl_artifacts:
        naming_checks[
            artifact_name
        ] = path.name.endswith(
            "_v1.jsonl"
        )

    elif artifact_name in json_report_artifacts:
        naming_checks[
            artifact_name
        ] = path.name.endswith(
            "_v1.json"
        )

    else:
        naming_checks[
            artifact_name
        ] = False

# -------------------------------------------------------------------
# Path collision checks
# -------------------------------------------------------------------

output_path_strings = [
    str(
        path.resolve()
    )
    for path
    in resolved_output_paths.values()
]

duplicate_output_paths = [
    path
    for path, count
    in Counter(
        output_path_strings
    ).items()
    if count > 1
]

all_paths_unique = (
    len(
        duplicate_output_paths
    )
    == 0
)

# -------------------------------------------------------------------
# Parent directory readiness
# -------------------------------------------------------------------

parent_directory_checks = {
    artifact_name: (
        path.parent.exists()
        and path.parent.is_dir()
    )
    for artifact_name, path
    in resolved_output_paths.items()
}

# -------------------------------------------------------------------
# Input contract
# -------------------------------------------------------------------

input_dataset_path = (
    repo_root
    / nb09_artifact_contract[
        "input"
    ][
        "normalized_dataset"
    ]
)

input_dataset_exists = (
    input_dataset_path.exists()
    and input_dataset_path.is_file()
)

# -------------------------------------------------------------------
# Expected in-memory populations
# -------------------------------------------------------------------

expected_artifact_populations = {
    "accepted_processing_dataset": (
        len(
            accepted_processing_records
        )
    ),
    "hard_quarantine_dataset": (
        len(
            hard_quarantine_records
        )
    ),
    "quality_evidence_records": (
        len(
            quality_policy_rows
        )
    ),
    "review_queue": (
        len(
            review_queue_records
        )
    ),
    "exact_duplicate_relations": (
        len(
            exact_duplicate_groups_final
        )
    ),
    "near_duplicate_candidates": (
        len(
            near_duplicate_candidate_relations
        )
    ),
    "cross_source_similarity_candidates": (
        len(
            cross_source_candidate_relations
        )
    ),
}

print(
    "\n=== EXPECTED ARTIFACT POPULATIONS ==="
)

for artifact_name, count in (
    expected_artifact_populations.items()
):
    print(
        "-",
        artifact_name,
        ":",
        count,
    )

expected_populations_valid = (
    expected_artifact_populations[
        "accepted_processing_dataset"
    ]
    == 19456
    and expected_artifact_populations[
        "hard_quarantine_dataset"
    ]
    == 0
    and expected_artifact_populations[
        "quality_evidence_records"
    ]
    == 19456
    and expected_artifact_populations[
        "review_queue"
    ]
    == 7
    and expected_artifact_populations[
        "exact_duplicate_relations"
    ]
    == 192
    and expected_artifact_populations[
        "near_duplicate_candidates"
    ]
    == 15929
    and expected_artifact_populations[
        "cross_source_similarity_candidates"
    ]
    == 37796
)

# -------------------------------------------------------------------
# Rerun state
# -------------------------------------------------------------------

existing_output_count = len(
    existing_output_files
)

if existing_output_count == 0:
    output_state = (
        "CLEAN_FIRST_BUILD"
    )

elif existing_output_count == len(
    resolved_output_paths
):
    output_state = (
        "COMPLETE_EXISTING_OUTPUT_SET_"
        "READY_FOR_IDEMPOTENT_REBUILD"
    )

else:
    output_state = (
        "PARTIAL_EXISTING_OUTPUT_SET_"
        "READY_FOR_RECOVERY_REBUILD"
    )

existing_outputs_safe = (
    len(
        invalid_existing_output_objects
    )
    == 0
)

print(
    "\n=== IDEMPOTENT REBUILD STATE ==="
)

print(
    "Existing planned output files:",
    existing_output_count,
)

print(
    "Output state:",
    output_state,
)

print(
    "Invalid existing output objects:",
    (
        invalid_existing_output_objects
        if invalid_existing_output_objects
        else "NONE"
    ),
)

# -------------------------------------------------------------------
# Final validation
# -------------------------------------------------------------------

all_paths_inside_allowed_roots = all(
    path_root_checks.values()
)

all_names_versioned = all(
    naming_checks.values()
)

all_parent_directories_ready = all(
    parent_directory_checks.values()
)

print(
    "\nValidation:"
)

checks = {
    "Normalized input artifact exists":
        input_dataset_exists,

    "All output paths are unique":
        all_paths_unique,

    "All outputs stay inside allowed Phase 2 roots":
        all_paths_inside_allowed_roots,

    "All artifact names follow discovered version convention":
        all_names_versioned,

    "All required parent directories already exist":
        all_parent_directories_ready,

    "Expected in-memory artifact populations reproduced":
        expected_populations_valid,

    "Existing planned outputs are safe for idempotent rebuild":
        existing_outputs_safe,

    "Rerun policy explicitly allows planned output replacement":
        (
            nb09_artifact_contract[
                "rerun_policy"
            ][
                "mode"
            ]
            == "IDEMPOTENT_REBUILD"
            and nb09_artifact_contract[
                "rerun_policy"
            ][
                "planned_existing_outputs_allowed"
            ]
            is True
        ),
}

for check_name, passed in (
    checks.items()
):
    print(
        check_name + ":",
        "PASS"
        if passed
        else "FAIL",
    )

artifact_contract_frozen = all(
    checks.values()
)

print(
    "CODE CELL 38 GATE:",
    "PASS"
    if artifact_contract_frozen
    else "FAIL",
)
```

```python
# Code Cell 39

from pathlib import Path
from collections import Counter
import hashlib
import json
import os
import tempfile

print(
    "=== NOTEBOOK 09 ARTIFACT BUILD ==="
)

# -------------------------------------------------------------------
# Helpers
# -------------------------------------------------------------------

def make_json_safe(value):
    if isinstance(value, dict):
        return {
            str(key): make_json_safe(item)
            for key, item in value.items()
        }

    if isinstance(value, (list, tuple)):
        return [
            make_json_safe(item)
            for item in value
        ]

    if isinstance(value, set):
        return sorted(
            make_json_safe(item)
            for item in value
        )

    if isinstance(value, Path):
        return str(value)

    if (
        hasattr(value, "item")
        and callable(getattr(value, "item"))
    ):
        try:
            return make_json_safe(
                value.item()
            )
        except (
            ValueError,
            TypeError,
        ):
            pass

    return value


def write_jsonl(path, records):
    path.parent.mkdir(
        parents=True,
        exist_ok=True,
    )

    with path.open(
        "w",
        encoding="utf-8",
        newline="\n",
    ) as handle:
        for record in records:
            handle.write(
                json.dumps(
                    make_json_safe(record),
                    ensure_ascii=False,
                    sort_keys=True,
                    separators=(",", ":"),
                )
            )
            handle.write("\n")


def write_json(path, payload):
    path.parent.mkdir(
        parents=True,
        exist_ok=True,
    )

    with path.open(
        "w",
        encoding="utf-8",
        newline="\n",
    ) as handle:
        json.dump(
            make_json_safe(payload),
            handle,
            ensure_ascii=False,
            sort_keys=True,
            indent=2,
        )
        handle.write("\n")


def sha256_file(path):
    digest = hashlib.sha256()

    with path.open("rb") as handle:
        for chunk in iter(
            lambda: handle.read(
                1024 * 1024
            ),
            b"",
        ):
            digest.update(chunk)

    return digest.hexdigest()


def jsonl_record_count(path):
    count = 0

    with path.open(
        "r",
        encoding="utf-8",
    ) as handle:
        for line_number, line in enumerate(
            handle,
            start=1,
        ):
            if not line.strip():
                continue

            json.loads(line)
            count += 1

    return count


def validate_jsonl_exact(
    path,
    expected_records,
):
    expected_count = len(
        expected_records
    )

    with path.open(
        "r",
        encoding="utf-8",
    ) as handle:
        for line_number, (
            line,
            expected_record,
        ) in enumerate(
            zip(
                handle,
                expected_records,
            ),
            start=1,
        ):
            parsed = json.loads(line)

            expected_safe = (
                make_json_safe(
                    expected_record
                )
            )

            if parsed != expected_safe:
                return (
                    False,
                    (
                        "CONTENT_MISMATCH_AT_LINE_"
                        + str(line_number)
                    ),
                )

    actual_count = (
        jsonl_record_count(path)
    )

    if actual_count != expected_count:
        return (
            False,
            (
                "COUNT_MISMATCH_"
                + str(actual_count)
                + "_EXPECTED_"
                + str(expected_count)
            ),
        )

    return True, "PASS"


def validate_json_exact(
    path,
    expected_payload,
):
    with path.open(
        "r",
        encoding="utf-8",
    ) as handle:
        parsed = json.load(handle)

    if (
        parsed
        != make_json_safe(
            expected_payload
        )
    ):
        return (
            False,
            "CONTENT_MISMATCH",
        )

    return True, "PASS"


def file_metadata(
    staged_path,
    final_relative_path,
    record_count=None,
):
    metadata = {
        "path": final_relative_path,
        "bytes": (
            staged_path.stat().st_size
        ),
        "sha256": (
            sha256_file(
                staged_path
            )
        ),
    }

    if record_count is not None:
        metadata[
            "record_count"
        ] = record_count

    return metadata


# -------------------------------------------------------------------
# Frozen final paths
# -------------------------------------------------------------------

final_paths = {
    artifact_name: (
        repo_root
        / relative_path
    )
    for artifact_name, relative_path
    in nb09_artifact_contract[
        "outputs"
    ].items()
}

# -------------------------------------------------------------------
# Existing-output safety for reruns
# -------------------------------------------------------------------

invalid_existing_objects = [
    str(
        path.relative_to(
            repo_root
        )
    )
    for path in final_paths.values()
    if (
        path.exists()
        and not path.is_file()
    )
]

if invalid_existing_objects:
    raise RuntimeError(
        "NB09 planned output path is not a file: "
        + repr(
            invalid_existing_objects
        )
    )

existing_before_build = {
    artifact_name: (
        path.exists()
    )
    for artifact_name, path
    in final_paths.items()
}

print(
    "Existing planned outputs before rebuild:",
    sum(
        existing_before_build.values()
    ),
)

print(
    "Build mode:",
    nb09_artifact_contract[
        "rerun_policy"
    ][
        "mode"
    ],
)

# -------------------------------------------------------------------
# JSONL payloads
# -------------------------------------------------------------------

jsonl_payloads = {
    "accepted_processing_dataset":
        accepted_processing_records,

    "hard_quarantine_dataset":
        hard_quarantine_records,

    "quality_evidence_records":
        quality_policy_rows,

    "review_queue":
        review_queue_records,

    "exact_duplicate_relations":
        exact_duplicate_groups_final,

    "near_duplicate_candidates":
        near_duplicate_candidate_relations,

    "cross_source_similarity_candidates":
        cross_source_candidate_relations,
}

expected_jsonl_counts = {
    artifact_name: len(records)
    for artifact_name, records
    in jsonl_payloads.items()
}

# -------------------------------------------------------------------
# Input metadata
# -------------------------------------------------------------------

normalized_input_path = (
    repo_root
    / nb09_artifact_contract[
        "input"
    ][
        "normalized_dataset"
    ]
)

normalized_input_metadata = {
    "path": str(
        normalized_input_path.relative_to(
            repo_root
        )
    ),
    "bytes": (
        normalized_input_path.stat().st_size
    ),
    "sha256": (
        sha256_file(
            normalized_input_path
        )
    ),
    "record_count": (
        len(quality_records)
    ),
}

# -------------------------------------------------------------------
# Report distributions
# -------------------------------------------------------------------

source_distribution = dict(
    Counter(
        record["source_type"]
        for record in quality_records
    )
)

quality_score_distribution = {
    str(score): count
    for score, count
    in sorted(
        Counter(
            row[
                "quality_completeness_score"
            ]
            for row
            in quality_policy_rows
        ).items()
    )
}

authority_distribution = dict(
    sorted(
        Counter(
            row["authority_class"]
            for row
            in quality_policy_rows
        ).items()
    )
)

rights_distribution = dict(
    sorted(
        Counter(
            row["rights_evidence_class"]
            for row
            in quality_policy_rows
        ).items()
    )
)

soft_metadata_distribution = dict(
    sorted(
        Counter(
            flag
            for row
            in quality_policy_rows
            for flag in row[
                "soft_metadata_flags"
            ]
        ).items()
    )
)

review_flag_distribution = dict(
    sorted(
        Counter(
            flag
            for row
            in quality_policy_rows
            for flag in row[
                "review_flags"
            ]
        ).items()
    )
)

near_band_distribution = dict(
    near_similarity_bands
)

cross_pair_distribution = {
    "|".join(pair): count
    for pair, count
    in sorted(
        cross_source_pair_type_counts.items()
    )
}

# -------------------------------------------------------------------
# Staging build
# -------------------------------------------------------------------

staging_validation = {}

with tempfile.TemporaryDirectory(
    prefix="nb09_staging_",
    dir=str(repo_root),
) as staging_root_raw:

    staging_root = Path(
        staging_root_raw
    )

    staged_paths = {
        artifact_name: (
            staging_root
            / final_path.name
        )
        for artifact_name, final_path
        in final_paths.items()
    }

    # ---------------------------------------------------------------
    # Seven JSONL artifacts
    # ---------------------------------------------------------------

    for artifact_name, records in (
        jsonl_payloads.items()
    ):
        write_jsonl(
            staged_paths[
                artifact_name
            ],
            records,
        )

    for artifact_name, records in (
        jsonl_payloads.items()
    ):
        passed, detail = (
            validate_jsonl_exact(
                staged_paths[
                    artifact_name
                ],
                records,
            )
        )

        staging_validation[
            artifact_name
        ] = {
            "passed": passed,
            "detail": detail,
        }

    jsonl_stage_pass = all(
        result["passed"]
        for result
        in staging_validation.values()
    )

    if not jsonl_stage_pass:
        raise RuntimeError(
            "NB09 JSONL staging validation failed: "
            + repr(
                staging_validation
            )
        )

    # ---------------------------------------------------------------
    # Seven data artifact metadata records
    # ---------------------------------------------------------------

    data_artifact_metadata = {}

    for artifact_name in (
        jsonl_payloads
    ):
        data_artifact_metadata[
            artifact_name
        ] = file_metadata(
            staged_paths[
                artifact_name
            ],
            nb09_artifact_contract[
                "outputs"
            ][
                artifact_name
            ],
            expected_jsonl_counts[
                artifact_name
            ],
        )

    # ---------------------------------------------------------------
    # Main report
    # ---------------------------------------------------------------

    dedup_quality_report = {
        "report_type": (
            "phase2_deduplication_quality_report"
        ),
        "report_version": "v1",
        "notebook": (
            "09_deduplication_and_quality.ipynb"
        ),
        "input": (
            normalized_input_metadata
        ),
        "policy": (
            nb09_policy_contract
        ),
        "artifact_contract": (
            nb09_artifact_contract
        ),
        "dataset": {
            "record_count": 19456,
            "source_distribution": (
                source_distribution
            ),
            "schema_valid": 19456,
            "schema_invalid": 0,
            "hard_structural_failures": 0,
        },
        "quality": {
            "score_name": (
                "structural_completeness_score_v1"
            ),
            "score_semantics": (
                "UNWEIGHTED_STRUCTURAL_COMPLETENESS_ONLY"
            ),
            "components": [
                "SCHEMA_VALID",
                "TEXT_NONEMPTY",
                "LOCATOR_INTEGRITY",
                "LINEAGE_INTEGRITY",
                "TEXT_HASH_INTEGRITY",
            ],
            "score_distribution": (
                quality_score_distribution
            ),
            "soft_metadata_flag_distribution": (
                soft_metadata_distribution
            ),
            "review_required_count": 7,
            "review_flag_distribution": (
                review_flag_distribution
            ),
        },
        "authority": {
            "primary_knowledge_authority": (
                "Dr. Komeil Roudi"
            ),
            "distribution": (
                authority_distribution
            ),
            "authority_is_quality_score": False,
            "human_confirmed_program_context": (
                human_confirmed_program_context
            ),
        },
        "rights": {
            "repository_policy": (
                "PROCESSING_ALLOWED_RELEASE_RIGHTS_GATED"
            ),
            "distribution": (
                rights_distribution
            ),
            "processing_allowed_count": 19456,
            "release_blocked_pending_rights_count": (
                19456
            ),
            "release_rights_gate_satisfied_count": (
                0
            ),
            "repository_presence_is_rights_evidence":
                False,
        },
        "deduplication": {
            "automatic_merge": False,
            "automatic_delete": False,
            "automatic_canonical_selection": False,
            "exact": {
                "group_count": (
                    exact_group_count
                ),
                "member_count": (
                    exact_member_count
                ),
                "relation_scope_distribution": (
                    dict(
                        exact_scope_counts
                    )
                ),
            },
            "near_duplicate": {
                "candidate_count": (
                    len(
                        near_duplicate_candidate_relations
                    )
                ),
                "source_variable": (
                    "pair_feature_records"
                ),
                "production_threshold_selected":
                    False,
                "similarity_band_distribution": (
                    near_band_distribution
                ),
            },
            "cross_source": {
                "candidate_count": (
                    len(
                        cross_source_candidate_relations
                    )
                ),
                "source_variable": (
                    "cross_pair_results"
                ),
                "production_threshold_selected":
                    False,
                "source_pair_distribution": (
                    cross_pair_distribution
                ),
                "similarity_is_duplicate_proof":
                    False,
            },
            "canonical_relation_policy": (
                "NO_AUTOMATIC_CANONICAL_DOCUMENT_WITHOUT_"
                "INDEPENDENT_EVIDENCE"
            ),
        },
        "processing": {
            "accepted_for_processing_count":
                19456,
            "hard_quarantine_count": 0,
            "review_queue_count": 7,
        },
        "release": {
            "release_dataset_created": False,
            "status": (
                "BLOCKED_PENDING_RIGHTS"
            ),
            "release_eligible_count": 0,
            "blocked_count": 19456,
            "quarantine_exclusion_assertion": (
                "NO_HARD_QUARANTINE_RECORD_EXISTS; "
                "ACCEPTED_AND_QUARANTINE_SETS_ARE_DISJOINT"
            ),
        },
        "known_limitations": [
            (
                "Rights evidence remains unresolved for "
                "all 19,456 documents for formal Release."
            ),
            (
                "Near-duplicate and cross-source similarity "
                "remain candidate relations; no production "
                "duplicate threshold was selected."
            ),
            (
                "Seven Khane-To transcript segments remain "
                "in human review due to unexpected participant labels."
            ),
            (
                "Asre-Shirin has human-confirmed two-season "
                "structure but season_number is absent in the "
                "current normalized dataset."
            ),
            (
                "2,354 transcript segments retain unresolved "
                "speaker context and are not guessed."
            ),
            (
                "Checkpoint artifact is deferred because no "
                "repository-local checkpoint directory/convention "
                "was present during discovery."
            ),
        ],
        "artifacts": (
            data_artifact_metadata
        ),
    }

    write_json(
        staged_paths[
            "dedup_quality_report"
        ],
        dedup_quality_report,
    )

    report_valid, report_detail = (
        validate_json_exact(
            staged_paths[
                "dedup_quality_report"
            ],
            dedup_quality_report,
        )
    )

    staging_validation[
        "dedup_quality_report"
    ] = {
        "passed": report_valid,
        "detail": report_detail,
    }

    if not report_valid:
        raise RuntimeError(
            "NB09 report staging validation failed: "
            + report_detail
        )

    report_metadata = (
        file_metadata(
            staged_paths[
                "dedup_quality_report"
            ],
            nb09_artifact_contract[
                "outputs"
            ][
                "dedup_quality_report"
            ],
        )
    )

    # ---------------------------------------------------------------
    # Acceptance
    # ---------------------------------------------------------------

    acceptance_checks = {
        "normalized_input_revalidated": (
            normalized_input_metadata[
                "record_count"
            ]
            == 19456
        ),
        "schema_invalid_zero": True,
        "hard_structural_failures_zero": (
            len(
                hard_quarantine_records
            )
            == 0
        ),
        "processing_population_complete": (
            len(
                accepted_processing_records
            )
            + len(
                hard_quarantine_records
            )
            == 19456
        ),
        "accepted_and_quarantine_disjoint": (
            accepted_processing_ids
            .isdisjoint(
                hard_quarantine_ids
            )
        ),
        "review_queue_count_is_7": (
            len(
                review_queue_records
            )
            == 7
        ),
        "exact_groups_are_192": (
            exact_group_count
            == 192
        ),
        "exact_members_are_931": (
            exact_member_count
            == 931
        ),
        "near_candidate_count_is_15929": (
            len(
                near_duplicate_candidate_relations
            )
            == 15929
        ),
        "cross_source_candidate_count_is_37796": (
            len(
                cross_source_candidate_relations
            )
            == 37796
        ),
        "automatic_merge_disabled": (
            nb09_policy_contract[
                "dedup"
            ][
                "automatic_merge"
            ]
            is False
        ),
        "automatic_delete_disabled": (
            nb09_policy_contract[
                "dedup"
            ][
                "automatic_delete"
            ]
            is False
        ),
        "automatic_canonicalization_zero": (
            all(
                group[
                    "canonical_document_id"
                ]
                is None
                for group
                in exact_duplicate_groups_final
            )
            and all(
                relation[
                    "canonical_document_id"
                ]
                is None
                for relation
                in near_duplicate_candidate_relations
            )
            and all(
                relation[
                    "canonical_document_id"
                ]
                is None
                for relation
                in cross_source_candidate_relations
            )
        ),
        "release_blocked_pending_rights_is_19456": (
            len(
                release_blocked_ids
            )
            == 19456
        ),
        "formal_release_eligible_count_is_zero": (
            len(
                release_rights_satisfied_ids
            )
            == 0
        ),
        "all_staged_data_artifacts_valid": (
            jsonl_stage_pass
        ),
        "dedup_quality_report_valid": (
            report_valid
        ),
        "rerun_policy_is_idempotent": (
            nb09_artifact_contract[
                "rerun_policy"
            ][
                "mode"
            ]
            == "IDEMPOTENT_REBUILD"
        ),
    }

    acceptance_status = (
        "PASS"
        if all(
            acceptance_checks.values()
        )
        else "FAIL"
    )

    dedup_quality_acceptance = {
        "report_type": (
            "phase2_deduplication_quality_acceptance"
        ),
        "report_version": "v1",
        "notebook": (
            "09_deduplication_and_quality.ipynb"
        ),
        "status": (
            acceptance_status
        ),
        "scope": (
            "NB09_PROCESSING_ACCEPTANCE_AND_"
            "DEDUP_QUALITY_COMPLETE"
        ),
        "processing_decision": {
            "status": "ACCEPTED",
            "record_count": 19456,
            "hard_quarantine_count": 0,
        },
        "release_decision": {
            "status": (
                "BLOCKED_PENDING_RIGHTS"
            ),
            "release_dataset_created": False,
            "eligible_count": 0,
            "blocked_count": 19456,
        },
        "review": {
            "required_count": 7,
            "automatic_quarantine": False,
        },
        "checks": (
            acceptance_checks
        ),
        "artifacts": {
            **data_artifact_metadata,
            "dedup_quality_report": (
                report_metadata
            ),
        },
        "open_gates": {
            "rights_evidence": {
                "status": (
                    "REQUIRED_FOR_FORMAL_RELEASE"
                ),
                "affected_document_count": (
                    19456
                ),
            },
            "checkpoint": {
                "status": (
                    "DEFERRED_NO_REPOSITORY_LOCAL_CONVENTION"
                ),
            },
        },
        "next_notebook": (
            "10_taxonomy_and_knowledge_engineering.ipynb"
        ),
        "next_notebook_allowed_for_internal_processing":
            (
                acceptance_status
                == "PASS"
            ),
        "constraints_for_next_notebook": [
            (
                "Use only the accepted processing dataset "
                "produced by Notebook 09."
            ),
            (
                "Preserve source identity, attribution, rights, "
                "locator, lineage, and dedup relations."
            ),
            (
                "Do not interpret duplicate relations as "
                "automatic source deletion or independent evidence."
            ),
            (
                "Do not treat formal Release as rights-cleared "
                "until rights evidence is supplied."
            ),
        ],
    }

    write_json(
        staged_paths[
            "dedup_quality_acceptance"
        ],
        dedup_quality_acceptance,
    )

    acceptance_valid, acceptance_detail = (
        validate_json_exact(
            staged_paths[
                "dedup_quality_acceptance"
            ],
            dedup_quality_acceptance,
        )
    )

    staging_validation[
        "dedup_quality_acceptance"
    ] = {
        "passed": acceptance_valid,
        "detail": acceptance_detail,
    }

    if not acceptance_valid:
        raise RuntimeError(
            "NB09 acceptance staging validation failed: "
            + acceptance_detail
        )

    # ---------------------------------------------------------------
    # Full staging gate
    # ---------------------------------------------------------------

    all_staged_artifacts_valid = all(
        result["passed"]
        for result
        in staging_validation.values()
    )

    if not all_staged_artifacts_valid:
        raise RuntimeError(
            "NB09 staging gate failed."
        )

    # ---------------------------------------------------------------
    # Idempotent publication:
    # validated staged files replace planned existing outputs.
    # ---------------------------------------------------------------

    publication_order = [
        "accepted_processing_dataset",
        "hard_quarantine_dataset",
        "quality_evidence_records",
        "review_queue",
        "exact_duplicate_relations",
        "near_duplicate_candidates",
        "cross_source_similarity_candidates",
        "dedup_quality_report",
        "dedup_quality_acceptance",
    ]

    for artifact_name in (
        publication_order
    ):
        final_path = (
            final_paths[
                artifact_name
            ]
        )

        final_path.parent.mkdir(
            parents=True,
            exist_ok=True,
        )

        os.replace(
            staged_paths[
                artifact_name
            ],
            final_path,
        )

# -------------------------------------------------------------------
# Independent post-publication validation
# -------------------------------------------------------------------

print(
    "\n=== POST-PUBLICATION ARTIFACT VALIDATION ==="
)

post_publication_checks = {}
final_artifact_metadata = {}

for artifact_name, records in (
    jsonl_payloads.items()
):
    final_path = (
        final_paths[
            artifact_name
        ]
    )

    exists = (
        final_path.exists()
        and final_path.is_file()
    )

    if exists:
        actual_count = (
            jsonl_record_count(
                final_path
            )
        )

        expected_count = len(
            records
        )

        count_match = (
            actual_count
            == expected_count
        )

        final_artifact_metadata[
            artifact_name
        ] = {
            "path": str(
                final_path.relative_to(
                    repo_root
                )
            ),
            "record_count": (
                actual_count
            ),
            "bytes": (
                final_path.stat().st_size
            ),
            "sha256": (
                sha256_file(
                    final_path
                )
            ),
        }

    else:
        count_match = False

    post_publication_checks[
        artifact_name
    ] = (
        exists
        and count_match
    )

report_path = (
    final_paths[
        "dedup_quality_report"
    ]
)

acceptance_path = (
    final_paths[
        "dedup_quality_acceptance"
    ]
)

report_post_valid = (
    report_path.exists()
    and report_path.is_file()
    and validate_json_exact(
        report_path,
        dedup_quality_report,
    )[0]
)

acceptance_post_valid = (
    acceptance_path.exists()
    and acceptance_path.is_file()
    and validate_json_exact(
        acceptance_path,
        dedup_quality_acceptance,
    )[0]
)

post_publication_checks[
    "dedup_quality_report"
] = report_post_valid

post_publication_checks[
    "dedup_quality_acceptance"
] = acceptance_post_valid

final_artifact_metadata[
    "dedup_quality_report"
] = {
    "path": str(
        report_path.relative_to(
            repo_root
        )
    ),
    "bytes": (
        report_path.stat().st_size
    ),
    "sha256": (
        sha256_file(
            report_path
        )
    ),
}

final_artifact_metadata[
    "dedup_quality_acceptance"
] = {
    "path": str(
        acceptance_path.relative_to(
            repo_root
        )
    ),
    "bytes": (
        acceptance_path.stat().st_size
    ),
    "sha256": (
        sha256_file(
            acceptance_path
        )
    ),
}

# -------------------------------------------------------------------
# On-disk identity checks
# -------------------------------------------------------------------

accepted_disk_ids = set()

with final_paths[
    "accepted_processing_dataset"
].open(
    "r",
    encoding="utf-8",
) as handle:
    for line in handle:
        if not line.strip():
            continue

        accepted_disk_ids.add(
            json.loads(
                line
            )[
                "document_id"
            ]
        )

quarantine_disk_ids = set()

with final_paths[
    "hard_quarantine_dataset"
].open(
    "r",
    encoding="utf-8",
) as handle:
    for line in handle:
        if not line.strip():
            continue

        quarantine_disk_ids.add(
            json.loads(
                line
            )[
                "document_id"
            ]
        )

accepted_population_disk_complete = (
    len(
        accepted_disk_ids
    )
    == 19456
)

quarantine_population_disk_zero = (
    len(
        quarantine_disk_ids
    )
    == 0
)

accepted_quarantine_disk_disjoint = (
    accepted_disk_ids.isdisjoint(
        quarantine_disk_ids
    )
)

all_final_artifacts_valid = all(
    post_publication_checks.values()
)

print(
    "Artifacts published:",
    len(
        final_paths
    ),
)

for artifact_name in (
    publication_order
):
    metadata = (
        final_artifact_metadata[
            artifact_name
        ]
    )

    print(
        "\n-",
        artifact_name,
    )

    print(
        "  path:",
        metadata["path"],
    )

    if (
        "record_count"
        in metadata
    ):
        print(
            "  records:",
            metadata[
                "record_count"
            ],
        )

    print(
        "  bytes:",
        metadata["bytes"],
    )

    print(
        "  sha256:",
        metadata["sha256"],
    )

print(
    "\nValidation:"
)

checks = {
    "All 9 final artifacts exist and validate":
        all_final_artifacts_valid,

    "Accepted dataset contains 19,456 unique document IDs":
        accepted_population_disk_complete,

    "Hard quarantine dataset contains zero document IDs":
        quarantine_population_disk_zero,

    "Accepted and quarantine datasets are disjoint on disk":
        accepted_quarantine_disk_disjoint,

    "Notebook acceptance report status is PASS":
        (
            dedup_quality_acceptance[
                "status"
            ]
            == "PASS"
        ),

    "Formal Release remains blocked pending rights":
        (
            dedup_quality_acceptance[
                "release_decision"
            ][
                "status"
            ]
            == "BLOCKED_PENDING_RIGHTS"
        ),

    "Notebook 10 allowed for internal processing":
        (
            dedup_quality_acceptance[
                "next_notebook_allowed_for_internal_processing"
            ]
            is True
        ),

    "Rebuild remained idempotent":
        (
            nb09_artifact_contract[
                "rerun_policy"
            ][
                "mode"
            ]
            == "IDEMPOTENT_REBUILD"
        ),
}

for check_name, passed in (
    checks.items()
):
    print(
        check_name + ":",
        "PASS"
        if passed
        else "FAIL",
    )

cell_gate = all(
    checks.values()
)

print(
    "CODE CELL 39 GATE:",
    "PASS"
    if cell_gate
    else "FAIL",
)
```

```python
# Code Cell 40

from pathlib import Path
from collections import Counter
import hashlib
import json

print(
    "=== NB09 INDEPENDENT DISK REVALIDATION ==="
)

# -------------------------------------------------------------------
# Paths defined independently from the build payload
# -------------------------------------------------------------------

disk_paths = {
    "input_normalized": (
        repo_root
        / "data/phase2/processed/"
          "documents_normalized_v1.jsonl"
    ),
    "accepted_processing_dataset": (
        repo_root
        / "data/phase2/processed/"
          "documents_dedup_quality_accepted_v1.jsonl"
    ),
    "hard_quarantine_dataset": (
        repo_root
        / "data/phase2/processed/"
          "documents_dedup_quality_quarantine_v1.jsonl"
    ),
    "quality_evidence_records": (
        repo_root
        / "data/phase2/processed/"
          "dedup_quality_evidence_records_v1.jsonl"
    ),
    "review_queue": (
        repo_root
        / "data/phase2/processed/"
          "dedup_quality_review_queue_v1.jsonl"
    ),
    "exact_duplicate_relations": (
        repo_root
        / "data/phase2/processed/"
          "dedup_exact_relations_v1.jsonl"
    ),
    "near_duplicate_candidates": (
        repo_root
        / "data/phase2/processed/"
          "dedup_near_candidates_v1.jsonl"
    ),
    "cross_source_similarity_candidates": (
        repo_root
        / "data/phase2/processed/"
          "dedup_cross_source_candidates_v1.jsonl"
    ),
    "dedup_quality_report": (
        repo_root
        / "reports/phase2/"
          "deduplication_quality_report_v1.json"
    ),
    "dedup_quality_acceptance": (
        repo_root
        / "reports/phase2/"
          "deduplication_quality_acceptance_v1.json"
    ),
}

for name, path in disk_paths.items():
    print(
        "-",
        name,
        "| exists:",
        path.exists(),
    )

# -------------------------------------------------------------------
# Helpers
# -------------------------------------------------------------------

def disk_sha256(path):
    digest = hashlib.sha256()

    with path.open(
        "rb"
    ) as handle:
        for chunk in iter(
            lambda: handle.read(
                1024 * 1024
            ),
            b"",
        ):
            digest.update(
                chunk
            )

    return digest.hexdigest()


def load_jsonl_from_disk(path):
    records = []

    with path.open(
        "r",
        encoding="utf-8",
    ) as handle:
        for line_number, line in enumerate(
            handle,
            start=1,
        ):
            if not line.strip():
                continue

            try:
                record = json.loads(
                    line
                )
            except json.JSONDecodeError as exc:
                raise RuntimeError(
                    f"Invalid JSONL at "
                    f"{path.name}:{line_number}"
                ) from exc

            records.append(
                record
            )

    return records


def load_json_from_disk(path):
    with path.open(
        "r",
        encoding="utf-8",
    ) as handle:
        return json.load(
            handle
        )


# -------------------------------------------------------------------
# Existence gate before reads
# -------------------------------------------------------------------

all_required_files_exist = all(
    path.exists()
    and path.is_file()
    for path in disk_paths.values()
)

if not all_required_files_exist:
    missing = [
        str(
            path.relative_to(
                repo_root
            )
        )
        for path in disk_paths.values()
        if (
            not path.exists()
            or not path.is_file()
        )
    ]

    raise RuntimeError(
        "Missing NB09 artifact(s): "
        + repr(
            missing
        )
    )

# -------------------------------------------------------------------
# Independent loads
# -------------------------------------------------------------------

input_disk_records = (
    load_jsonl_from_disk(
        disk_paths[
            "input_normalized"
        ]
    )
)

accepted_disk_records = (
    load_jsonl_from_disk(
        disk_paths[
            "accepted_processing_dataset"
        ]
    )
)

quarantine_disk_records = (
    load_jsonl_from_disk(
        disk_paths[
            "hard_quarantine_dataset"
        ]
    )
)

quality_evidence_disk_records = (
    load_jsonl_from_disk(
        disk_paths[
            "quality_evidence_records"
        ]
    )
)

review_disk_records = (
    load_jsonl_from_disk(
        disk_paths[
            "review_queue"
        ]
    )
)

exact_disk_relations = (
    load_jsonl_from_disk(
        disk_paths[
            "exact_duplicate_relations"
        ]
    )
)

near_disk_relations = (
    load_jsonl_from_disk(
        disk_paths[
            "near_duplicate_candidates"
        ]
    )
)

cross_disk_relations = (
    load_jsonl_from_disk(
        disk_paths[
            "cross_source_similarity_candidates"
        ]
    )
)

report_disk = (
    load_json_from_disk(
        disk_paths[
            "dedup_quality_report"
        ]
    )
)

acceptance_disk = (
    load_json_from_disk(
        disk_paths[
            "dedup_quality_acceptance"
        ]
    )
)

# -------------------------------------------------------------------
# Independent counts
# -------------------------------------------------------------------

observed_counts = {
    "input_normalized": (
        len(
            input_disk_records
        )
    ),
    "accepted_processing_dataset": (
        len(
            accepted_disk_records
        )
    ),
    "hard_quarantine_dataset": (
        len(
            quarantine_disk_records
        )
    ),
    "quality_evidence_records": (
        len(
            quality_evidence_disk_records
        )
    ),
    "review_queue": (
        len(
            review_disk_records
        )
    ),
    "exact_duplicate_relations": (
        len(
            exact_disk_relations
        )
    ),
    "near_duplicate_candidates": (
        len(
            near_disk_relations
        )
    ),
    "cross_source_similarity_candidates": (
        len(
            cross_disk_relations
        )
    ),
}

expected_counts = {
    "input_normalized": 19456,
    "accepted_processing_dataset": 19456,
    "hard_quarantine_dataset": 0,
    "quality_evidence_records": 19456,
    "review_queue": 7,
    "exact_duplicate_relations": 192,
    "near_duplicate_candidates": 15929,
    "cross_source_similarity_candidates": 37796,
}

print(
    "\n=== INDEPENDENT RECORD COUNTS ==="
)

for name in expected_counts:
    print(
        "-",
        name,
        "| observed:",
        observed_counts[
            name
        ],
        "| expected:",
        expected_counts[
            name
        ],
    )

counts_match = (
    observed_counts
    == expected_counts
)

# -------------------------------------------------------------------
# Independent hashes
# -------------------------------------------------------------------

disk_hashes = {
    name: disk_sha256(
        path
    )
    for name, path
    in disk_paths.items()
}

print(
    "\n=== INDEPENDENT SHA-256 ==="
)

for name, value in (
    disk_hashes.items()
):
    print(
        "-",
        name,
        ":",
        value,
    )

# Because no record was quarantined or mutated,
# the accepted processing dataset must be byte-identical
# to the normalized input artifact.

accepted_matches_input_hash = (
    disk_hashes[
        "accepted_processing_dataset"
    ]
    == disk_hashes[
        "input_normalized"
    ]
)

# -------------------------------------------------------------------
# Document identity sets
# -------------------------------------------------------------------

input_ids = {
    record[
        "document_id"
    ]
    for record in input_disk_records
}

accepted_ids = {
    record[
        "document_id"
    ]
    for record in accepted_disk_records
}

quarantine_ids = {
    record[
        "document_id"
    ]
    for record in quarantine_disk_records
}

quality_evidence_ids = {
    record[
        "document_id"
    ]
    for record
    in quality_evidence_disk_records
}

review_ids = {
    record[
        "document_id"
    ]
    for record in review_disk_records
}

input_ids_unique = (
    len(
        input_ids
    )
    == 19456
)

accepted_ids_unique = (
    len(
        accepted_ids
    )
    == 19456
)

accepted_identity_matches_input = (
    accepted_ids
    == input_ids
)

quality_evidence_identity_matches_input = (
    quality_evidence_ids
    == input_ids
)

quarantine_empty = (
    len(
        quarantine_ids
    )
    == 0
)

accepted_quarantine_disjoint = (
    accepted_ids.isdisjoint(
        quarantine_ids
    )
)

review_subset_accepted = (
    review_ids.issubset(
        accepted_ids
    )
    and len(
        review_ids
    )
    == 7
)

# -------------------------------------------------------------------
# Quality evidence revalidation
# -------------------------------------------------------------------

quality_score_counts_disk = Counter(
    record[
        "quality_completeness_score"
    ]
    for record
    in quality_evidence_disk_records
)

processing_decision_counts_disk = Counter(
    record[
        "processing_decision"
    ]
    for record
    in quality_evidence_disk_records
)

release_eligibility_counts_disk = Counter(
    record[
        "release_eligibility"
    ]
    for record
    in quality_evidence_disk_records
)

quality_scores_expected = (
    quality_score_counts_disk
    == Counter(
        {
            1.0: 19456,
        }
    )
)

processing_decisions_expected = (
    processing_decision_counts_disk
    == Counter(
        {
            "ACCEPTED_FOR_PROCESSING": 19456,
        }
    )
)

release_decisions_expected = (
    release_eligibility_counts_disk
    == Counter(
        {
            "RELEASE_BLOCKED_PENDING_RIGHTS": 19456,
        }
    )
)

# -------------------------------------------------------------------
# Exact relation integrity
# -------------------------------------------------------------------

exact_member_references = []

exact_relation_ids = []

exact_self_group_failures = []

exact_canonical_nonnull = []

for relation in exact_disk_relations:
    relation_id = relation[
        "relation_group_id"
    ]

    member_ids = relation[
        "member_document_ids"
    ]

    exact_relation_ids.append(
        relation_id
    )

    exact_member_references.extend(
        member_ids
    )

    if (
        len(
            member_ids
        )
        < 2
        or len(
            member_ids
        )
        != len(
            set(
                member_ids
            )
        )
    ):
        exact_self_group_failures.append(
            relation_id
        )

    if (
        relation.get(
            "canonical_document_id"
        )
        is not None
    ):
        exact_canonical_nonnull.append(
            relation_id
        )

exact_all_members_exist = all(
    document_id
    in accepted_ids
    for document_id
    in exact_member_references
)

exact_member_count_disk = len(
    exact_member_references
)

exact_relation_ids_unique = (
    len(
        exact_relation_ids
    )
    == len(
        set(
            exact_relation_ids
        )
    )
)

exact_group_structure_valid = (
    len(
        exact_self_group_failures
    )
    == 0
)

exact_no_auto_canonical = (
    len(
        exact_canonical_nonnull
    )
    == 0
)

exact_scope_counts_disk = Counter(
    relation[
        "relation_scope"
    ]
    for relation
    in exact_disk_relations
)

exact_scope_expected = (
    exact_scope_counts_disk
    == Counter(
        {
            "SAME_SOURCE_ID": 91,
            "SAME_SOURCE_TYPE_DIFFERENT_SOURCE_ID": 92,
            "CROSS_SOURCE_TYPE": 9,
        }
    )
)

# -------------------------------------------------------------------
# Pair relation integrity helper
# -------------------------------------------------------------------

def validate_pair_relations(
    relations,
    valid_document_ids,
):
    relation_ids = []
    missing_endpoint_count = 0
    self_pair_count = 0
    canonical_nonnull_count = 0

    for relation in relations:
        relation_id = relation[
            "relation_id"
        ]

        document_id_a = relation[
            "document_id_a"
        ]

        document_id_b = relation[
            "document_id_b"
        ]

        relation_ids.append(
            relation_id
        )

        if (
            document_id_a
            not in valid_document_ids
            or document_id_b
            not in valid_document_ids
        ):
            missing_endpoint_count += 1

        if (
            document_id_a
            == document_id_b
        ):
            self_pair_count += 1

        if (
            relation.get(
                "canonical_document_id"
            )
            is not None
        ):
            canonical_nonnull_count += 1

    return {
        "relation_ids_unique": (
            len(
                relation_ids
            )
            == len(
                set(
                    relation_ids
                )
            )
        ),
        "missing_endpoint_count": (
            missing_endpoint_count
        ),
        "self_pair_count": (
            self_pair_count
        ),
        "canonical_nonnull_count": (
            canonical_nonnull_count
        ),
    }


near_integrity = (
    validate_pair_relations(
        near_disk_relations,
        accepted_ids,
    )
)

cross_integrity = (
    validate_pair_relations(
        cross_disk_relations,
        accepted_ids,
    )
)

near_relations_valid = (
    near_integrity[
        "relation_ids_unique"
    ]
    and near_integrity[
        "missing_endpoint_count"
    ]
    == 0
    and near_integrity[
        "self_pair_count"
    ]
    == 0
    and near_integrity[
        "canonical_nonnull_count"
    ]
    == 0
)

cross_relations_valid = (
    cross_integrity[
        "relation_ids_unique"
    ]
    and cross_integrity[
        "missing_endpoint_count"
    ]
    == 0
    and cross_integrity[
        "self_pair_count"
    ]
    == 0
    and cross_integrity[
        "canonical_nonnull_count"
    ]
    == 0
)

# -------------------------------------------------------------------
# Reproduce near similarity bands from disk
# -------------------------------------------------------------------

def disk_similarity_band(
    score,
):
    score = float(
        score
    )

    if score >= 0.98:
        return "GE_0_98"

    if score >= 0.95:
        return "0_95_TO_0_98"

    if score >= 0.90:
        return "0_90_TO_0_95"

    if score >= 0.80:
        return "0_80_TO_0_90"

    return "LT_0_80"


near_bands_disk = Counter(
    disk_similarity_band(
        relation[
            "char_tfidf_cosine"
        ]
    )
    for relation
    in near_disk_relations
)

near_bands_expected = (
    near_bands_disk
    == Counter(
        {
            "GE_0_98": 52,
            "0_95_TO_0_98": 46,
            "0_90_TO_0_95": 96,
            "0_80_TO_0_90": 461,
            "LT_0_80": 15274,
        }
    )
)

# -------------------------------------------------------------------
# Reproduce cross-source pair distribution from disk
# -------------------------------------------------------------------

cross_pair_distribution_disk = Counter(
    tuple(
        sorted(
            (
                relation[
                    "source_type_a"
                ],
                relation[
                    "source_type_b"
                ],
            )
        )
    )
    for relation
    in cross_disk_relations
)

cross_pair_distribution_expected = (
    cross_pair_distribution_disk
    == Counter(
        {
            (
                "book",
                "transcript",
            ): 18146,
            (
                "book",
                "website",
            ): 3096,
            (
                "transcript",
                "website",
            ): 16554,
        }
    )
)

# -------------------------------------------------------------------
# Report artifact metadata consistency
# -------------------------------------------------------------------

report_artifacts = (
    report_disk[
        "artifacts"
    ]
)

report_artifact_checks = {}

for artifact_name in [
    "accepted_processing_dataset",
    "hard_quarantine_dataset",
    "quality_evidence_records",
    "review_queue",
    "exact_duplicate_relations",
    "near_duplicate_candidates",
    "cross_source_similarity_candidates",
]:
    metadata = (
        report_artifacts[
            artifact_name
        ]
    )

    actual_path = (
        disk_paths[
            artifact_name
        ]
    )

    report_artifact_checks[
        artifact_name
    ] = (
        metadata[
            "sha256"
        ]
        == disk_hashes[
            artifact_name
        ]
        and metadata[
            "bytes"
        ]
        == actual_path.stat().st_size
        and metadata[
            "record_count"
        ]
        == observed_counts[
            artifact_name
        ]
    )

report_metadata_consistent = all(
    report_artifact_checks.values()
)

# -------------------------------------------------------------------
# Acceptance metadata consistency
# -------------------------------------------------------------------

acceptance_artifacts = (
    acceptance_disk[
        "artifacts"
    ]
)

acceptance_artifact_checks = {}

for artifact_name in [
    "accepted_processing_dataset",
    "hard_quarantine_dataset",
    "quality_evidence_records",
    "review_queue",
    "exact_duplicate_relations",
    "near_duplicate_candidates",
    "cross_source_similarity_candidates",
]:
    metadata = (
        acceptance_artifacts[
            artifact_name
        ]
    )

    actual_path = (
        disk_paths[
            artifact_name
        ]
    )

    acceptance_artifact_checks[
        artifact_name
    ] = (
        metadata[
            "sha256"
        ]
        == disk_hashes[
            artifact_name
        ]
        and metadata[
            "bytes"
        ]
        == actual_path.stat().st_size
        and metadata[
            "record_count"
        ]
        == observed_counts[
            artifact_name
        ]
    )

acceptance_report_metadata = (
    acceptance_artifacts[
        "dedup_quality_report"
    ]
)

acceptance_report_hash_consistent = (
    acceptance_report_metadata[
        "sha256"
    ]
    == disk_hashes[
        "dedup_quality_report"
    ]
    and acceptance_report_metadata[
        "bytes"
    ]
    == disk_paths[
        "dedup_quality_report"
    ].stat().st_size
)

acceptance_metadata_consistent = (
    all(
        acceptance_artifact_checks.values()
    )
    and acceptance_report_hash_consistent
)

# -------------------------------------------------------------------
# Report / Acceptance semantic contract
# -------------------------------------------------------------------

report_semantics_valid = (
    report_disk[
        "dataset"
    ][
        "record_count"
    ]
    == 19456
    and report_disk[
        "processing"
    ][
        "accepted_for_processing_count"
    ]
    == 19456
    and report_disk[
        "processing"
    ][
        "hard_quarantine_count"
    ]
    == 0
    and report_disk[
        "processing"
    ][
        "review_queue_count"
    ]
    == 7
    and report_disk[
        "release"
    ][
        "status"
    ]
    == "BLOCKED_PENDING_RIGHTS"
    and report_disk[
        "release"
    ][
        "release_eligible_count"
    ]
    == 0
    and report_disk[
        "deduplication"
    ][
        "exact"
    ][
        "group_count"
    ]
    == 192
    and report_disk[
        "deduplication"
    ][
        "exact"
    ][
        "member_count"
    ]
    == 931
    and report_disk[
        "deduplication"
    ][
        "near_duplicate"
    ][
        "candidate_count"
    ]
    == 15929
    and report_disk[
        "deduplication"
    ][
        "cross_source"
    ][
        "candidate_count"
    ]
    == 37796
)

acceptance_semantics_valid = (
    acceptance_disk[
        "status"
    ]
    == "PASS"
    and acceptance_disk[
        "processing_decision"
    ][
        "record_count"
    ]
    == 19456
    and acceptance_disk[
        "processing_decision"
    ][
        "hard_quarantine_count"
    ]
    == 0
    and acceptance_disk[
        "review"
    ][
        "required_count"
    ]
    == 7
    and acceptance_disk[
        "release_decision"
    ][
        "status"
    ]
    == "BLOCKED_PENDING_RIGHTS"
    and acceptance_disk[
        "release_decision"
    ][
        "eligible_count"
    ]
    == 0
    and acceptance_disk[
        "release_decision"
    ][
        "blocked_count"
    ]
    == 19456
    and acceptance_disk[
        "next_notebook_allowed_for_internal_processing"
    ]
    is True
)

# -------------------------------------------------------------------
# Final independent gate
# -------------------------------------------------------------------

print(
    "\n=== RELATION INTEGRITY ==="
)

print(
    "Exact member references:",
    exact_member_count_disk,
)

print(
    "Exact relation IDs unique:",
    exact_relation_ids_unique,
)

print(
    "Near integrity:",
    near_integrity,
)

print(
    "Cross-source integrity:",
    cross_integrity,
)

print(
    "\nValidation:"
)

checks = {
    "All 10 input/output files exist":
        all_required_files_exist,

    "All expected record counts match":
        counts_match,

    "Accepted artifact is byte-identical to normalized input":
        accepted_matches_input_hash,

    "Input document IDs are unique":
        input_ids_unique,

    "Accepted document IDs are unique":
        accepted_ids_unique,

    "Accepted identity set equals normalized input":
        accepted_identity_matches_input,

    "Quality evidence covers exactly the accepted identity set":
        quality_evidence_identity_matches_input,

    "Hard quarantine is truly empty":
        quarantine_empty,

    "Accepted and quarantine are disjoint":
        accepted_quarantine_disjoint,

    "Review queue is exactly 7 accepted documents":
        review_subset_accepted,

    "Quality score distribution is 1.0 for all 19,456":
        quality_scores_expected,

    "Processing decision distribution is correct":
        processing_decisions_expected,

    "Release eligibility distribution is correct":
        release_decisions_expected,

    "All exact relation members exist in accepted dataset":
        exact_all_members_exist,

    "Exact relation member count is 931":
        (
            exact_member_count_disk
            == 931
        ),

    "Exact relation IDs are unique":
        exact_relation_ids_unique,

    "Exact group member structure is valid":
        exact_group_structure_valid,

    "Exact relation scopes reproduce expected counts":
        exact_scope_expected,

    "Exact relations have no automatic canonical":
        exact_no_auto_canonical,

    "Near-duplicate relation integrity is clean":
        near_relations_valid,

    "Near similarity bands reproduce expected counts":
        near_bands_expected,

    "Cross-source relation integrity is clean":
        cross_relations_valid,

    "Cross-source pair distribution reproduces expected counts":
        cross_pair_distribution_expected,

    "Main report metadata matches disk artifacts":
        report_metadata_consistent,

    "Acceptance metadata matches disk artifacts":
        acceptance_metadata_consistent,

    "Main report semantic contract is valid":
        report_semantics_valid,

    "Acceptance semantic contract is valid":
        acceptance_semantics_valid,
}

for check_name, passed in (
    checks.items()
):
    print(
        check_name + ":",
        "PASS"
        if passed
        else "FAIL",
    )

cell_gate = all(
    checks.values()
)

print(
    "CODE CELL 40 GATE:",
    "PASS"
    if cell_gate
    else "FAIL",
)
```

```python
# Code Cell 41

from pathlib import Path
import hashlib
import json
import os
import tempfile

print(
    "=== NB09 CHECKPOINT BUILD ==="
)

# -------------------------------------------------------------------
# Checkpoint contract
# -------------------------------------------------------------------

checkpoint_relative_path = (
    "artifacts/phase2/checkpoints/"
    "09_deduplication_and_quality.json"
)

checkpoint_path = (
    repo_root
    / checkpoint_relative_path
)

checkpoint_root = (
    checkpoint_path.parent
)

print(
    "Checkpoint path:",
    checkpoint_relative_path,
)

print(
    "Checkpoint directory exists before build:",
    checkpoint_root.exists(),
)

print(
    "Checkpoint file exists before build:",
    checkpoint_path.exists(),
)

# -------------------------------------------------------------------
# Safety
# -------------------------------------------------------------------

if (
    checkpoint_path.exists()
    and not checkpoint_path.is_file()
):
    raise RuntimeError(
        "Checkpoint path exists but is not a file."
    )

checkpoint_root.mkdir(
    parents=True,
    exist_ok=True,
)

# -------------------------------------------------------------------
# Independent helpers
# -------------------------------------------------------------------

def checkpoint_sha256(
    path,
):
    digest = hashlib.sha256()

    with path.open(
        "rb"
    ) as handle:
        for chunk in iter(
            lambda: handle.read(
                1024 * 1024
            ),
            b"",
        ):
            digest.update(
                chunk
            )

    return digest.hexdigest()


def checkpoint_json_load(
    path,
):
    with path.open(
        "r",
        encoding="utf-8",
    ) as handle:
        return json.load(
            handle
        )


def checkpoint_json_write(
    path,
    payload,
):
    with path.open(
        "w",
        encoding="utf-8",
        newline="\n",
    ) as handle:
        json.dump(
            payload,
            handle,
            ensure_ascii=False,
            sort_keys=True,
            indent=2,
        )

        handle.write(
            "\n"
        )


# -------------------------------------------------------------------
# Re-read current final artifacts directly from disk
# -------------------------------------------------------------------

checkpoint_artifact_paths = {
    "accepted_processing_dataset": (
        repo_root
        / "data/phase2/processed/"
          "documents_dedup_quality_accepted_v1.jsonl"
    ),
    "hard_quarantine_dataset": (
        repo_root
        / "data/phase2/processed/"
          "documents_dedup_quality_quarantine_v1.jsonl"
    ),
    "quality_evidence_records": (
        repo_root
        / "data/phase2/processed/"
          "dedup_quality_evidence_records_v1.jsonl"
    ),
    "review_queue": (
        repo_root
        / "data/phase2/processed/"
          "dedup_quality_review_queue_v1.jsonl"
    ),
    "exact_duplicate_relations": (
        repo_root
        / "data/phase2/processed/"
          "dedup_exact_relations_v1.jsonl"
    ),
    "near_duplicate_candidates": (
        repo_root
        / "data/phase2/processed/"
          "dedup_near_candidates_v1.jsonl"
    ),
    "cross_source_similarity_candidates": (
        repo_root
        / "data/phase2/processed/"
          "dedup_cross_source_candidates_v1.jsonl"
    ),
    "dedup_quality_report": (
        repo_root
        / "reports/phase2/"
          "deduplication_quality_report_v1.json"
    ),
    "dedup_quality_acceptance": (
        repo_root
        / "reports/phase2/"
          "deduplication_quality_acceptance_v1.json"
    ),
}

all_checkpoint_artifacts_exist = all(
    path.exists()
    and path.is_file()
    for path
    in checkpoint_artifact_paths.values()
)

if not all_checkpoint_artifacts_exist:
    missing = [
        str(
            path.relative_to(
                repo_root
            )
        )
        for path
        in checkpoint_artifact_paths.values()
        if (
            not path.exists()
            or not path.is_file()
        )
    ]

    raise RuntimeError(
        "Missing NB09 artifact before checkpoint build: "
        + repr(
            missing
        )
    )

# -------------------------------------------------------------------
# Load acceptance + report independently
# -------------------------------------------------------------------

acceptance_disk_for_checkpoint = (
    checkpoint_json_load(
        checkpoint_artifact_paths[
            "dedup_quality_acceptance"
        ]
    )
)

report_disk_for_checkpoint = (
    checkpoint_json_load(
        checkpoint_artifact_paths[
            "dedup_quality_report"
        ]
    )
)

# -------------------------------------------------------------------
# Final artifact metadata
# -------------------------------------------------------------------

checkpoint_artifact_metadata = {}

for artifact_name, path in (
    checkpoint_artifact_paths.items()
):
    metadata = {
        "path": str(
            path.relative_to(
                repo_root
            )
        ),
        "bytes": (
            path.stat().st_size
        ),
        "sha256": (
            checkpoint_sha256(
                path
            )
        ),
    }

    if path.suffix == ".jsonl":
        record_count = 0

        with path.open(
            "r",
            encoding="utf-8",
        ) as handle:
            for line in handle:
                if not line.strip():
                    continue

                json.loads(
                    line
                )

                record_count += 1

        metadata[
            "record_count"
        ] = record_count

    checkpoint_artifact_metadata[
        artifact_name
    ] = metadata

# -------------------------------------------------------------------
# Input metadata
# -------------------------------------------------------------------

checkpoint_input_path = (
    repo_root
    / "data/phase2/processed/"
      "documents_normalized_v1.jsonl"
)

checkpoint_input_metadata = {
    "path": str(
        checkpoint_input_path.relative_to(
            repo_root
        )
    ),
    "bytes": (
        checkpoint_input_path.stat().st_size
    ),
    "sha256": (
        checkpoint_sha256(
            checkpoint_input_path
        )
    ),
    "record_count": 19456,
}

# -------------------------------------------------------------------
# Checkpoint payload
# -------------------------------------------------------------------

nb09_checkpoint = {
    "checkpoint_type": (
        "phase2_notebook_checkpoint"
    ),
    "checkpoint_version": (
        "v1"
    ),
    "notebook_number": 9,
    "notebook": (
        "09_deduplication_and_quality.ipynb"
    ),
    "status": (
        "PASS"
    ),
    "scope": (
        "NB09_COMPLETE_NB10_INTERNAL_PROCESSING_ALLOWED"
    ),
    "input": (
        checkpoint_input_metadata
    ),
    "processing": {
        "accepted_count": 19456,
        "hard_quarantine_count": 0,
        "review_queue_count": 7,
    },
    "deduplication": {
        "exact_group_count": 192,
        "exact_member_count": 931,
        "near_duplicate_candidate_count": 15929,
        "cross_source_similarity_candidate_count": 37796,
        "automatic_merge": False,
        "automatic_delete": False,
        "automatic_canonical_selection": False,
    },
    "quality": {
        "structural_completeness_score_name": (
            "structural_completeness_score_v1"
        ),
        "all_documents_structurally_complete": True,
        "hard_failure_count": 0,
    },
    "release": {
        "status": (
            "BLOCKED_PENDING_RIGHTS"
        ),
        "release_eligible_count": 0,
        "blocked_count": 19456,
        "formal_release_created": False,
    },
    "next_notebook": (
        "10_taxonomy_and_knowledge_engineering.ipynb"
    ),
    "next_notebook_allowed_for_internal_processing": True,
    "artifacts": (
        checkpoint_artifact_metadata
    ),
    "evidence": {
        "acceptance_report_status": (
            acceptance_disk_for_checkpoint[
                "status"
            ]
        ),
        "report_release_status": (
            report_disk_for_checkpoint[
                "release"
            ][
                "status"
            ]
        ),
    },
    "open_gates": {
        "rights_evidence": (
            "REQUIRED_FOR_FORMAL_RELEASE"
        ),
    },
}

# -------------------------------------------------------------------
# Staging write + exact validation
# -------------------------------------------------------------------

with tempfile.TemporaryDirectory(
    prefix="nb09_checkpoint_",
    dir=str(
        repo_root
    ),
) as staging_root_raw:

    staging_root = Path(
        staging_root_raw
    )

    staged_checkpoint = (
        staging_root
        / checkpoint_path.name
    )

    checkpoint_json_write(
        staged_checkpoint,
        nb09_checkpoint,
    )

    staged_loaded = (
        checkpoint_json_load(
            staged_checkpoint
        )
    )

    staged_content_valid = (
        staged_loaded
        == nb09_checkpoint
    )

    if not staged_content_valid:
        raise RuntimeError(
            "NB09 checkpoint staging content mismatch."
        )

    os.replace(
        staged_checkpoint,
        checkpoint_path,
    )

# -------------------------------------------------------------------
# Independent post-publication validation
# -------------------------------------------------------------------

checkpoint_disk = (
    checkpoint_json_load(
        checkpoint_path
    )
)

checkpoint_content_valid = (
    checkpoint_disk
    == nb09_checkpoint
)

checkpoint_status_valid = (
    checkpoint_disk[
        "status"
    ]
    == "PASS"
)

checkpoint_processing_valid = (
    checkpoint_disk[
        "processing"
    ][
        "accepted_count"
    ]
    == 19456
    and checkpoint_disk[
        "processing"
    ][
        "hard_quarantine_count"
    ]
    == 0
    and checkpoint_disk[
        "processing"
    ][
        "review_queue_count"
    ]
    == 7
)

checkpoint_dedup_valid = (
    checkpoint_disk[
        "deduplication"
    ][
        "exact_group_count"
    ]
    == 192
    and checkpoint_disk[
        "deduplication"
    ][
        "exact_member_count"
    ]
    == 931
    and checkpoint_disk[
        "deduplication"
    ][
        "near_duplicate_candidate_count"
    ]
    == 15929
    and checkpoint_disk[
        "deduplication"
    ][
        "cross_source_similarity_candidate_count"
    ]
    == 37796
)

checkpoint_release_valid = (
    checkpoint_disk[
        "release"
    ][
        "status"
    ]
    == "BLOCKED_PENDING_RIGHTS"
    and checkpoint_disk[
        "release"
    ][
        "release_eligible_count"
    ]
    == 0
    and checkpoint_disk[
        "release"
    ][
        "blocked_count"
    ]
    == 19456
)

checkpoint_next_notebook_valid = (
    checkpoint_disk[
        "next_notebook_allowed_for_internal_processing"
    ]
    is True
)

checkpoint_artifact_hashes_valid = all(
    checkpoint_disk[
        "artifacts"
    ][
        artifact_name
    ][
        "sha256"
    ]
    == checkpoint_sha256(
        path
    )
    for artifact_name, path
    in checkpoint_artifact_paths.items()
)

acceptance_semantics_valid = (
    acceptance_disk_for_checkpoint[
        "status"
    ]
    == "PASS"
    and acceptance_disk_for_checkpoint[
        "release_decision"
    ][
        "status"
    ]
    == "BLOCKED_PENDING_RIGHTS"
)

report_semantics_valid = (
    report_disk_for_checkpoint[
        "release"
    ][
        "status"
    ]
    == "BLOCKED_PENDING_RIGHTS"
)

print(
    "\n=== CHECKPOINT ARTIFACT ==="
)

print(
    "path:",
    checkpoint_relative_path,
)

print(
    "bytes:",
    checkpoint_path.stat().st_size,
)

print(
    "sha256:",
    checkpoint_sha256(
        checkpoint_path
    ),
)

print(
    "\nValidation:"
)

checks = {
    "All referenced NB09 artifacts exist":
        all_checkpoint_artifacts_exist,

    "Checkpoint content round-trip is exact":
        checkpoint_content_valid,

    "Checkpoint status is PASS":
        checkpoint_status_valid,

    "Processing population is correct":
        checkpoint_processing_valid,

    "Dedup populations are correct":
        checkpoint_dedup_valid,

    "Formal Release remains blocked pending rights":
        checkpoint_release_valid,

    "Notebook 10 internal processing is allowed":
        checkpoint_next_notebook_valid,

    "All artifact hashes in checkpoint match disk":
        checkpoint_artifact_hashes_valid,

    "Acceptance semantics match checkpoint":
        acceptance_semantics_valid,

    "Main report semantics match checkpoint":
        report_semantics_valid,
}

for check_name, passed in (
    checks.items()
):
    print(
        check_name + ":",
        "PASS"
        if passed
        else "FAIL",
    )

cell_gate = all(
    checks.values()
)

print(
    "CODE CELL 41 GATE:",
    "PASS"
    if cell_gate
    else "FAIL",
)
```

<li>وجود 192 گروه Exact Duplicate با 931 عضو.</li>
<li>وجود 15,929 Near-Duplicate Candidate.</li>
<li>وجود 37,796 Cross-source Similarity Candidate.</li>
<li>PASS بودن Acceptance Report.</li>
<li>PASS بودن Checkpoint.</li>
<li>تطابق Hash تمام Artifactهای ثبت‌شده در Checkpoint با فایل واقعی.</li>
<li>مجاز بودن ادامه Processing داخلی در Notebook 10.</li>
<li>باقی‌ماندن Release رسمی در وضعیت Blocked Pending Rights.</li>
</ul>

<h4>معیار نهایی</h4>

<p>
فقط در صورتی که تمام کنترل‌ها PASS باشند،
Notebook 09 از نظر Final Health آماده آزمون
<code>Restart Kernel → Run All</code>
خواهد بود.

```python
# Code Cell 42

from pathlib import Path
import hashlib
import json

print(
    "=== NOTEBOOK 09 FINAL HEALTH AUDIT ==="
)

# -------------------------------------------------------------------
# Independent paths
# -------------------------------------------------------------------

health_paths = {
    "input_normalized": (
        repo_root
        / "data/phase2/processed/"
          "documents_normalized_v1.jsonl"
    ),
    "accepted_processing_dataset": (
        repo_root
        / "data/phase2/processed/"
          "documents_dedup_quality_accepted_v1.jsonl"
    ),
    "hard_quarantine_dataset": (
        repo_root
        / "data/phase2/processed/"
          "documents_dedup_quality_quarantine_v1.jsonl"
    ),
    "quality_evidence_records": (
        repo_root
        / "data/phase2/processed/"
          "dedup_quality_evidence_records_v1.jsonl"
    ),
    "review_queue": (
        repo_root
        / "data/phase2/processed/"
          "dedup_quality_review_queue_v1.jsonl"
    ),
    "exact_duplicate_relations": (
        repo_root
        / "data/phase2/processed/"
          "dedup_exact_relations_v1.jsonl"
    ),
    "near_duplicate_candidates": (
        repo_root
        / "data/phase2/processed/"
          "dedup_near_candidates_v1.jsonl"
    ),
    "cross_source_similarity_candidates": (
        repo_root
        / "data/phase2/processed/"
          "dedup_cross_source_candidates_v1.jsonl"
    ),
    "dedup_quality_report": (
        repo_root
        / "reports/phase2/"
          "deduplication_quality_report_v1.json"
    ),
    "dedup_quality_acceptance": (
        repo_root
        / "reports/phase2/"
          "deduplication_quality_acceptance_v1.json"
    ),
    "checkpoint": (
        repo_root
        / "artifacts/phase2/checkpoints/"
          "09_deduplication_and_quality.json"
    ),
}

# -------------------------------------------------------------------
# Helpers
# -------------------------------------------------------------------

def health_sha256(
    path,
):
    digest = hashlib.sha256()

    with path.open(
        "rb"
    ) as handle:
        for chunk in iter(
            lambda: handle.read(
                1024 * 1024
            ),
            b"",
        ):
            digest.update(
                chunk
            )

    return digest.hexdigest()


def health_jsonl_count(
    path,
):
    count = 0

    with path.open(
        "r",
        encoding="utf-8",
    ) as handle:
        for line_number, line in enumerate(
            handle,
            start=1,
        ):
            if not line.strip():
                continue

            json.loads(
                line
            )

            count += 1

    return count


def health_load_json(
    path,
):
    with path.open(
        "r",
        encoding="utf-8",
    ) as handle:
        return json.load(
            handle
        )


# -------------------------------------------------------------------
# File existence
# -------------------------------------------------------------------

all_files_exist = all(
    path.exists()
    and path.is_file()
    for path in health_paths.values()
)

print(
    "Required files:",
    len(
        health_paths
    ),
)

print(
    "All required files exist:",
    all_files_exist,
)

if not all_files_exist:
    missing = [
        str(
            path.relative_to(
                repo_root
            )
        )
        for path in health_paths.values()
        if (
            not path.exists()
            or not path.is_file()
        )
    ]

    raise RuntimeError(
        "Missing final NB09 file(s): "
        + repr(
            missing
        )
    )

# -------------------------------------------------------------------
# Record counts
# -------------------------------------------------------------------

expected_counts = {
    "input_normalized": 19456,
    "accepted_processing_dataset": 19456,
    "hard_quarantine_dataset": 0,
    "quality_evidence_records": 19456,
    "review_queue": 7,
    "exact_duplicate_relations": 192,
    "near_duplicate_candidates": 15929,
    "cross_source_similarity_candidates": 37796,
}

observed_counts = {
    name: health_jsonl_count(
        health_paths[
            name
        ]
    )
    for name in expected_counts
}

counts_valid = (
    observed_counts
    == expected_counts
)

print(
    "\n=== FINAL RECORD COUNTS ==="
)

for name in expected_counts:
    print(
        "-",
        name,
        "| observed:",
        observed_counts[
            name
        ],
        "| expected:",
        expected_counts[
            name
        ],
    )

# -------------------------------------------------------------------
# Hashes
# -------------------------------------------------------------------

health_hashes = {
    name: health_sha256(
        path
    )
    for name, path
    in health_paths.items()
}

accepted_equals_normalized = (
    health_hashes[
        "accepted_processing_dataset"
    ]
    == health_hashes[
        "input_normalized"
    ]
)

print(
    "\n=== FINAL HASH CONTRACT ==="
)

print(
    "Normalized input SHA-256:",
    health_hashes[
        "input_normalized"
    ],
)

print(
    "Accepted dataset SHA-256:",
    health_hashes[
        "accepted_processing_dataset"
    ],
)

print(
    "Accepted is byte-identical to normalized input:",
    accepted_equals_normalized,
)

# -------------------------------------------------------------------
# Load final reports
# -------------------------------------------------------------------

report_health = (
    health_load_json(
        health_paths[
            "dedup_quality_report"
        ]
    )
)

acceptance_health = (
    health_load_json(
        health_paths[
            "dedup_quality_acceptance"
        ]
    )
)

checkpoint_health = (
    health_load_json(
        health_paths[
            "checkpoint"
        ]
    )
)

# -------------------------------------------------------------------
# Acceptance contract
# -------------------------------------------------------------------

acceptance_pass = (
    acceptance_health[
        "status"
    ]
    == "PASS"
)

processing_complete = (
    acceptance_health[
        "processing_decision"
    ][
        "status"
    ]
    == "ACCEPTED"
    and acceptance_health[
        "processing_decision"
    ][
        "record_count"
    ]
    == 19456
    and acceptance_health[
        "processing_decision"
    ][
        "hard_quarantine_count"
    ]
    == 0
)

review_valid = (
    acceptance_health[
        "review"
    ][
        "required_count"
    ]
    == 7
)

release_gate_valid = (
    acceptance_health[
        "release_decision"
    ][
        "status"
    ]
    == "BLOCKED_PENDING_RIGHTS"
    and acceptance_health[
        "release_decision"
    ][
        "eligible_count"
    ]
    == 0
    and acceptance_health[
        "release_decision"
    ][
        "blocked_count"
    ]
    == 19456
)

next_notebook_allowed = (
    acceptance_health[
        "next_notebook_allowed_for_internal_processing"
    ]
    is True
)

# -------------------------------------------------------------------
# Main report contract
# -------------------------------------------------------------------

report_valid = (
    report_health[
        "dataset"
    ][
        "record_count"
    ]
    == 19456
    and report_health[
        "processing"
    ][
        "accepted_for_processing_count"
    ]
    == 19456
    and report_health[
        "processing"
    ][
        "hard_quarantine_count"
    ]
    == 0
    and report_health[
        "processing"
    ][
        "review_queue_count"
    ]
    == 7
    and report_health[
        "deduplication"
    ][
        "exact"
    ][
        "group_count"
    ]
    == 192
    and report_health[
        "deduplication"
    ][
        "exact"
    ][
        "member_count"
    ]
    == 931
    and report_health[
        "deduplication"
    ][
        "near_duplicate"
    ][
        "candidate_count"
    ]
    == 15929
    and report_health[
        "deduplication"
    ][
        "cross_source"
    ][
        "candidate_count"
    ]
    == 37796
    and report_health[
        "release"
    ][
        "status"
    ]
    == "BLOCKED_PENDING_RIGHTS"
)

# -------------------------------------------------------------------
# Checkpoint contract
# -------------------------------------------------------------------

checkpoint_pass = (
    checkpoint_health[
        "status"
    ]
    == "PASS"
)

checkpoint_processing_valid = (
    checkpoint_health[
        "processing"
    ][
        "accepted_count"
    ]
    == 19456
    and checkpoint_health[
        "processing"
    ][
        "hard_quarantine_count"
    ]
    == 0
    and checkpoint_health[
        "processing"
    ][
        "review_queue_count"
    ]
    == 7
)

checkpoint_dedup_valid = (
    checkpoint_health[
        "deduplication"
    ][
        "exact_group_count"
    ]
    == 192
    and checkpoint_health[
        "deduplication"
    ][
        "exact_member_count"
    ]
    == 931
    and checkpoint_health[
        "deduplication"
    ][
        "near_duplicate_candidate_count"
    ]
    == 15929
    and checkpoint_health[
        "deduplication"
    ][
        "cross_source_similarity_candidate_count"
    ]
    == 37796
)

checkpoint_release_valid = (
    checkpoint_health[
        "release"
    ][
        "status"
    ]
    == "BLOCKED_PENDING_RIGHTS"
    and checkpoint_health[
        "release"
    ][
        "release_eligible_count"
    ]
    == 0
    and checkpoint_health[
        "release"
    ][
        "blocked_count"
    ]
    == 19456
)

checkpoint_next_notebook_valid = (
    checkpoint_health[
        "next_notebook_allowed_for_internal_processing"
    ]
    is True
)

# -------------------------------------------------------------------
# Checkpoint artifact hash validation
# -------------------------------------------------------------------

checkpoint_hash_mismatches = []

for artifact_name, metadata in (
    checkpoint_health[
        "artifacts"
    ].items()
):
    if artifact_name not in health_paths:
        checkpoint_hash_mismatches.append(
            (
                artifact_name,
                "PATH_NOT_IN_HEALTH_CONTRACT",
            )
        )
        continue

    actual_hash = health_hashes[
        artifact_name
    ]

    if (
        metadata[
            "sha256"
        ]
        != actual_hash
    ):
        checkpoint_hash_mismatches.append(
            (
                artifact_name,
                metadata[
                    "sha256"
                ],
                actual_hash,
            )
        )

checkpoint_hashes_valid = (
    len(
        checkpoint_hash_mismatches
    )
    == 0
)

# -------------------------------------------------------------------
# Report / acceptance linkage
# -------------------------------------------------------------------

acceptance_report_hash_valid = (
    acceptance_health[
        "artifacts"
    ][
        "dedup_quality_report"
    ][
        "sha256"
    ]
    == health_hashes[
        "dedup_quality_report"
    ]
)

# -------------------------------------------------------------------
# Hard quarantine real state
# -------------------------------------------------------------------

hard_quarantine_truly_empty = (
    health_paths[
        "hard_quarantine_dataset"
    ].stat().st_size
    == 0
    and observed_counts[
        "hard_quarantine_dataset"
    ]
    == 0
)

# -------------------------------------------------------------------
# Final Health Gate
# -------------------------------------------------------------------

print(
    "\nValidation:"
)

checks = {
    "All required final files exist":
        all_files_exist,

    "All final JSONL record counts match":
        counts_valid,

    "Accepted dataset equals normalized input byte-for-byte":
        accepted_equals_normalized,

    "Hard quarantine is truly empty":
        hard_quarantine_truly_empty,

    "Acceptance report status is PASS":
        acceptance_pass,

    "Processing acceptance is complete":
        processing_complete,

    "Review queue count is 7":
        review_valid,

    "Formal Release remains blocked pending rights":
        release_gate_valid,

    "Main report semantic contract is valid":
        report_valid,

    "Checkpoint status is PASS":
        checkpoint_pass,

    "Checkpoint processing contract is valid":
        checkpoint_processing_valid,

    "Checkpoint dedup contract is valid":
        checkpoint_dedup_valid,

    "Checkpoint release contract is valid":
        checkpoint_release_valid,

    "Checkpoint artifact hashes match current disk":
        checkpoint_hashes_valid,

    "Acceptance report references current report hash":
        acceptance_report_hash_valid,

    "Notebook 10 is allowed for internal processing":
        (
            next_notebook_allowed
            and checkpoint_next_notebook_valid
        ),
}

for check_name, passed in (
    checks.items()
):
    print(
        check_name + ":",
        "PASS"
        if passed
        else "FAIL",
    )

final_health_pass = all(
    checks.values()
)

print(
    "\nFINAL HEALTH:",
    "PASS"
    if final_health_pass
    else "FAIL",
)

print(
    "CODE CELL 42 GATE:",
    "PASS"
    if final_health_pass
    else "FAIL",
)
```

```python
# Code Cell 43

from pathlib import Path
from collections import Counter
import json
import re

print("=== NOTEBOOK 09 STRUCTURE HEALTH ===")

notebook_path = (
    repo_root
    / "notebooks/phase2/09_deduplication_and_quality.ipynb"
)

print(
    "Notebook:",
    notebook_path.relative_to(repo_root),
)

print(
    "Notebook exists:",
    notebook_path.exists(),
)

if not notebook_path.exists():
    raise RuntimeError(
        "Notebook file was not found on disk."
    )

with notebook_path.open(
    "r",
    encoding="utf-8",
) as handle:
    notebook_data = json.load(handle)

cells = notebook_data.get(
    "cells",
    [],
)

code_cells = [
    cell
    for cell in cells
    if cell.get("cell_type") == "code"
]

markdown_cells = [
    cell
    for cell in cells
    if cell.get("cell_type") == "markdown"
]

# ---------------------------------------------------------------
# Code-cell numbering
# ---------------------------------------------------------------

number_pattern = re.compile(
    r"^\s*#\s*Code Cell\s+(\d+)\s*$",
    re.IGNORECASE,
)

numbered_cells = []

unnumbered_code_cell_indexes = []

for code_index, cell in enumerate(
    code_cells,
    start=1,
):
    source = "".join(
        cell.get(
            "source",
            [],
        )
    )

    first_nonempty_line = None

    for line in source.splitlines():
        if line.strip():
            first_nonempty_line = (
                line.strip()
            )
            break

    match = (
        number_pattern.match(
            first_nonempty_line
        )
        if first_nonempty_line
        else None
    )

    if match:
        numbered_cells.append(
            {
                "position": code_index,
                "number": int(
                    match.group(1)
                ),
            }
        )
    else:
        unnumbered_code_cell_indexes.append(
            code_index
        )

numbers = [
    item["number"]
    for item in numbered_cells
]

number_counts = Counter(
    numbers
)

duplicate_numbers = sorted(
    number
    for number, count
    in number_counts.items()
    if count > 1
)

expected_numbers = list(
    range(
        1,
        len(code_cells) + 1,
    )
)

missing_numbers = sorted(
    set(expected_numbers)
    - set(numbers)
)

unexpected_numbers = sorted(
    set(numbers)
    - set(expected_numbers)
)

numbers_in_notebook_order = (
    numbers
)

sequential_numbering = (
    numbers_in_notebook_order
    == expected_numbers
)

# ---------------------------------------------------------------
# Execution status
# ---------------------------------------------------------------

executed_cells = 0
unexecuted_positions = []

for code_index, cell in enumerate(
    code_cells,
    start=1,
):
    execution_count = (
        cell.get(
            "execution_count"
        )
    )

    if execution_count is None:
        unexecuted_positions.append(
            code_index
        )
    else:
        executed_cells += 1

# Code Cell 43 is currently running.
# The saved notebook may still show its execution_count as None.
current_cell_is_only_unexecuted = (
    len(code_cells) == 43
    and unexecuted_positions
    in (
        [],
        [43],
    )
)

effective_executed_cells = (
    len(code_cells)
    if current_cell_is_only_unexecuted
    else executed_cells
)

effective_unexecuted_count = (
    0
    if current_cell_is_only_unexecuted
    else len(
        unexecuted_positions
    )
)

# ---------------------------------------------------------------
# Error outputs
# ---------------------------------------------------------------

error_cells = []

for code_index, cell in enumerate(
    code_cells,
    start=1,
):
    outputs = cell.get(
        "outputs",
        [],
    )

    has_error = any(
        output.get(
            "output_type"
        )
        == "error"
        for output in outputs
    )

    if has_error:
        error_cells.append(
            code_index
        )

# ---------------------------------------------------------------
# Final markers from actual saved outputs
# ---------------------------------------------------------------

saved_output_text_parts = []

for cell in code_cells:
    for output in cell.get(
        "outputs",
        [],
    ):
        output_type = output.get(
            "output_type"
        )

        if output_type == "stream":
            text_value = output.get(
                "text",
                "",
            )

            if isinstance(
                text_value,
                list,
            ):
                saved_output_text_parts.extend(
                    text_value
                )
            else:
                saved_output_text_parts.append(
                    str(
                        text_value
                    )
                )

        elif output_type in {
            "execute_result",
            "display_data",
        }:
            text_value = (
                output.get(
                    "data",
                    {},
                ).get(
                    "text/plain",
                    ""
                )
            )

            if isinstance(
                text_value,
                list,
            ):
                saved_output_text_parts.extend(
                    text_value
                )
            elif text_value:
                saved_output_text_parts.append(
                    str(
                        text_value
                    )
                )

saved_output_text = "".join(
    saved_output_text_parts
)

final_acceptance_marker = (
    "CODE CELL 42 GATE: PASS"
    in saved_output_text
)

final_health_marker = (
    "FINAL HEALTH: PASS"
    in saved_output_text
)

# ---------------------------------------------------------------
# Final artifact / acceptance / checkpoint status
# ---------------------------------------------------------------

acceptance_path = (
    repo_root
    / "reports/phase2/"
      "deduplication_quality_acceptance_v1.json"
)

checkpoint_path = (
    repo_root
    / "artifacts/phase2/checkpoints/"
      "09_deduplication_and_quality.json"
)

with acceptance_path.open(
    "r",
    encoding="utf-8",
) as handle:
    acceptance = json.load(
        handle
    )

with checkpoint_path.open(
    "r",
    encoding="utf-8",
) as handle:
    checkpoint = json.load(
        handle
    )

acceptance_pass = (
    acceptance.get(
        "status"
    )
    == "PASS"
)

checkpoint_pass = (
    checkpoint.get(
        "status"
    )
    == "PASS"
)

next_notebook_allowed = (
    acceptance.get(
        "next_notebook_allowed_for_internal_processing"
    )
    is True
    and checkpoint.get(
        "next_notebook_allowed_for_internal_processing"
    )
    is True
)

# ---------------------------------------------------------------
# Output
# ---------------------------------------------------------------

print(
    "\n=== CELL COUNTS ==="
)

print(
    "Markdown cells:",
    len(markdown_cells),
)

print(
    "Code cells:",
    len(code_cells),
)

print(
    "Executed:",
    effective_executed_cells,
)

print(
    "Unexecuted:",
    effective_unexecuted_count,
)

print(
    "Error cells:",
    len(error_cells),
)

print(
    "\n=== CODE CELL NUMBERING ==="
)

print(
    "Numbered cells:",
    f"{len(numbered_cells)}/{len(code_cells)}",
)

print(
    "Missing code-cell numbers:",
    len(missing_numbers),
)

print(
    "Missing numbers:",
    (
        missing_numbers
        if missing_numbers
        else "NONE"
    ),
)

print(
    "Duplicate code-cell numbers:",
    len(duplicate_numbers),
)

print(
    "Duplicate numbers:",
    (
        duplicate_numbers
        if duplicate_numbers
        else "NONE"
    ),
)

print(
    "Unexpected numbers:",
    (
        unexpected_numbers
        if unexpected_numbers
        else "NONE"
    ),
)

print(
    "Unnumbered code-cell positions:",
    (
        unnumbered_code_cell_indexes
        if unnumbered_code_cell_indexes
        else "NONE"
    ),
)

print(
    "Sequential numbering:",
    (
        "PASS"
        if sequential_numbering
        else "FAIL"
    ),
)

print(
    "\n=== FINAL MARKERS ==="
)

print(
    "Final acceptance marker:",
    (
        "PASS"
        if final_acceptance_marker
        else "FAIL"
    ),
)

print(
    "Final health marker:",
    (
        "PASS"
        if final_health_marker
        else "FAIL"
    ),
)

print(
    "Acceptance report:",
    (
        "PASS"
        if acceptance_pass
        else "FAIL"
    ),
)

print(
    "Checkpoint:",
    (
        "PASS"
        if checkpoint_pass
        else "FAIL"
    ),
)

print(
    "Notebook 10 internal processing allowed:",
    (
        "PASS"
        if next_notebook_allowed
        else "FAIL"
    ),
)

# ---------------------------------------------------------------
# Gate
# ---------------------------------------------------------------

checks = {
    "Exactly 43 code cells":
        len(code_cells) == 43,

    "All code cells numbered":
        len(numbered_cells)
        == len(code_cells),

    "No missing code-cell numbers":
        len(missing_numbers) == 0,

    "No duplicate code-cell numbers":
        len(duplicate_numbers) == 0,

    "No unexpected code-cell numbers":
        len(unexpected_numbers) == 0,

    "Sequential numbering":
        sequential_numbering,

    "All code cells effectively executed":
        effective_unexecuted_count == 0,

    "No saved error outputs":
        len(error_cells) == 0,

    "Final acceptance marker exists":
        final_acceptance_marker,

    "Final health marker exists":
        final_health_marker,

    "Acceptance report PASS":
        acceptance_pass,

    "Checkpoint PASS":
        checkpoint_pass,

    "Notebook 10 internal processing allowed":
        next_notebook_allowed,
}

print(
    "\n=== NOTEBOOK HEALTH CHECKS ==="
)

for check_name, passed in checks.items():
    print(
        check_name + ":",
        (
            "PASS"
            if passed
            else "FAIL"
        ),
    )

notebook_health_pass = all(
    checks.values()
)

print(
    "\nNOTEBOOK HEALTH:",
    (
        "PASS"
        if notebook_health_pass
        else "FAIL"
    ),
)

print(
    "CODE CELL 43 GATE:",
    (
        "PASS"
        if notebook_health_pass
        else "FAIL"
    ),
)
```
