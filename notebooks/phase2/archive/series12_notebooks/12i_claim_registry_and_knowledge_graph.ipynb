<h2>Repository و Artifact Preflight</h2>

<p>

این مرحله کاملاً Read-Only است و هیچ Artifact یا فایل موجودی را تغییر نمی‌دهد.

<p>

<ul>

<li>Path واقعی <code>12i_claim_registry_and_knowledge_graph.ipynb</code>؛</li>
<li>Notebookهای مرتبط قبلی، به‌خصوص NB12h و Notebookهای Benchmark؛</li>
<li>Accepted Corpus Candidateهای موجود؛</li>
<li>Taxonomy، Competency و Dedup Artifactهای مرتبط؛</li>

<li>اندازه و Record Count کم‌هزینهٔ Artifactهای حساس در صورت امکان؛</li>

<p>
اگر Source of Truth مبهم، چندگانه یا مفقود باشد، این مرحله PASS نخواهد شد و قبل از ادامه همان Preflight Repair خواهد شد.

```python
# Code Cell 01

from pathlib import Path
from collections import Counter, defaultdict
from concurrent.futures import (
    ThreadPoolExecutor,
    as_completed,
)

import hashlib
import json
import os
import re
import subprocess
import threading
import time

import numpy as np

from openai import OpenAI
```

```python
# Code Cell 02

def run_git(args, cwd):
    result = subprocess.run(
        ["git", *args],
        cwd=cwd,
        capture_output=True,
        text=True,
        check=False,
    )
    return result.returncode, result.stdout.strip(), result.stderr.strip()


def find_project_root(start_path):
    start_path = start_path.resolve()

    code, stdout, _ = run_git(
        ["rev-parse", "--show-toplevel"],
        start_path,
    )

    if code == 0 and stdout:
        return Path(stdout).resolve()

    for candidate in [start_path, *start_path.parents]:
        if (candidate / ".git").exists():
            return candidate.resolve()

    return None


def count_jsonl_records(path):
    count = 0

    with path.open(
        "r",
        encoding="utf-8",
        errors="strict",
    ) as file:
        for line in file:
            if line.strip():
                count += 1

    return count


def file_sha256(path):
    digest = hashlib.sha256()

    with path.open("rb") as file:
        for block in iter(
            lambda: file.read(1024 * 1024),
            b"",
        ):
            digest.update(block)

    return digest.hexdigest()


cwd = Path.cwd().resolve()
project_root = find_project_root(cwd)

assert project_root is not None, (
    "Project root could not be resolved from the current working directory."
)


git_branch_code, git_branch, git_branch_error = run_git(
    ["branch", "--show-current"],
    project_root,
)

git_status_code, git_status, git_status_error = run_git(
    ["status", "--short"],
    project_root,
)


current_notebook_matches = sorted(
    project_root.rglob(
        "12i_claim_registry_and_knowledge_graph.ipynb"
    )
)

related_notebooks = sorted(
    path
    for path in project_root.rglob("*.ipynb")
    if any(
        token in path.name.casefold()
        for token in [
            "12h",
            "12i",
            "13",
            "golden",
            "claim",
            "benchmark",
        ]
    )
)


accepted_corpus_matches = sorted(
    project_root.rglob(
        "documents_dedup_quality_accepted_v1.jsonl"
    )
)


relevant_artifact_names = [
    "taxonomy_v1.json",
    "competency_prerequisite_v1.json",
    "claim_discovery_population_index_v1.jsonl",
    "claim_discovery_population_prep_v1.json",
    "claim_discovery_dedup_topology_audit_v1.json",
    "claim_discovery_chunk_independence_audit_v1.json",
    "claim_extraction_structure_contract_audit_v1.json",
]


relevant_artifacts = {}

for artifact_name in relevant_artifact_names:
    matches = sorted(
        project_root.rglob(
            artifact_name
        )
    )

    relevant_artifacts[
        artifact_name
    ] = matches


claim_registry_patterns = [
    "claims_raw*.jsonl",
    "claims_validated*.jsonl",
    "claims_deduplicated*.jsonl",
    "claim_evidence*.jsonl",
    "claim_relations*.jsonl",
    "knowledge_graph*.json",
    "claim_registry_report*.json",
]


existing_registry_artifacts = []

for pattern in claim_registry_patterns:
    existing_registry_artifacts.extend(
        project_root.rglob(
            pattern
        )
    )

existing_registry_artifacts = sorted(
    set(
        existing_registry_artifacts
    )
)


notebook_profile = None

if len(
    current_notebook_matches
) == 1:

    notebook_path = (
        current_notebook_matches[0]
    )

    with notebook_path.open(
        "r",
        encoding="utf-8",
        errors="strict",
    ) as file:
        notebook_json = json.load(
            file
        )

    cell_types = Counter(
        cell.get("cell_type")
        for cell in notebook_json.get(
            "cells",
            [],
        )
    )

    notebook_profile = {
        "path": str(
            notebook_path.relative_to(
                project_root
            )
        ),
        "total_cells": len(
            notebook_json.get(
                "cells",
                []
            )
        ),
        "cell_types": dict(
            cell_types
        ),
    }


accepted_corpus_profile = None

if len(
    accepted_corpus_matches
) == 1:

    accepted_corpus_path = (
        accepted_corpus_matches[0]
    )

    accepted_corpus_profile = {
        "path": str(
            accepted_corpus_path.relative_to(
                project_root
            )
        ),
        "size_bytes": (
            accepted_corpus_path.stat().st_size
        ),
        "records": (
            count_jsonl_records(
                accepted_corpus_path
            )
        ),
        "sha256": (
            file_sha256(
                accepted_corpus_path
            )
        ),
    }


preflight_failures = []

if git_branch_code != 0:
    preflight_failures.append(
        "Git branch could not be resolved."
    )

if len(
    current_notebook_matches
) != 1:
    preflight_failures.append(
        "Current notebook was not found exactly once."
    )

if len(
    accepted_corpus_matches
) != 1:
    preflight_failures.append(
        "Accepted corpus candidate was not found exactly once."
    )


print(
    "=== NB12i REPOSITORY / ARTIFACT PREFLIGHT ==="
)

print()
print(
    "Project root:",
    project_root,
)

print(
    "Current working directory:",
    cwd,
)

print(
    "Git branch:",
    git_branch if git_branch_code == 0 else "UNRESOLVED",
)

print(
    "Git working tree entries:",
    len(
        git_status.splitlines()
    ) if git_status else 0,
)


print()
print(
    "Current notebook matches:",
    len(
        current_notebook_matches
    ),
)

for path in current_notebook_matches:
    print(
        "  -",
        path.relative_to(
            project_root
        ),
    )

if notebook_profile is not None:
    print(
        "Notebook profile:",
        notebook_profile,
    )


print()
print(
    "Related notebooks:",
    len(
        related_notebooks
    ),
)

for path in related_notebooks:
    print(
        "  -",
        path.relative_to(
            project_root
        ),
    )


print()
print(
    "Accepted corpus matches:",
    len(
        accepted_corpus_matches
    ),
)

if accepted_corpus_profile is not None:
    print(
        "Accepted corpus path:",
        accepted_corpus_profile[
            "path"
        ],
    )
    print(
        "Accepted corpus size bytes:",
        accepted_corpus_profile[
            "size_bytes"
        ],
    )
    print(
        "Accepted corpus records:",
        accepted_corpus_profile[
            "records"
        ],
    )
    print(
        "Accepted corpus SHA-256:",
        accepted_corpus_profile[
            "sha256"
        ],
    )


print()
print(
    "Relevant prerequisite artifacts:"
)

for artifact_name, matches in (
    relevant_artifacts.items()
):
    print(
        f"  - {artifact_name}:",
        len(matches),
    )

    for path in matches:
        print(
            "      ",
            path.relative_to(
                project_root
            ),
        )


print()
print(
    "Existing Claim Registry / "
    "Knowledge Graph artifacts:",
    len(
        existing_registry_artifacts
    ),
)

for path in (
    existing_registry_artifacts
):
    print(
        "  -",
        path.relative_to(
            project_root
        ),
    )


print()
print(
    "Preflight failures:",
    len(
        preflight_failures
    ),
)

for failure in preflight_failures:
    print(
        "  -",
        failure,
    )


preflight_status = (
    "PASS"
    if not preflight_failures
    else "FAIL"
)

print()
print(
    "CODE CELL 02 GATE:",
    preflight_status,
)
```

<h2>تثبیت Input Contract و Population رسمی Claim Registry</h2>

<p>
Preflight مرحلهٔ قبل نشان داد که Source Corpus پذیرفته‌شده، Taxonomy، Competency و Artifactهای لازم از NB12h همگی در Repository موجود هستند و Claim Registry یا Knowledge Graph قبلی نیز وجود ندارد.
</p>

<p>
در این مرحله باید دقیقاً مشخص شود Notebook جدید از کدام Artifactها به‌عنوان Source of Truth استفاده می‌کند و چه بخش‌هایی از Corpus اجازهٔ تولید Claim مستقل دارند.
</p>

<h3>اصل مهم</h3>

<p>
هدف این Notebook ساخت یک Claim Registry گسترده برای استفادهٔ مشترک در RAG، Golden Dataset و Book Production است؛ اما «گسترده بودن Registry» به معنی تبدیل تمام ۱۹٬۴۵۶ Processing Record به Claim معتبر نیست.
</p>

<p>
Authority Policy قبلی باید حفظ شود.
بنابراین Processing Recordها به چهار نقش تقسیم می‌شوند:
</p>

<ul>
<li><strong>Primary Claim Discovery:</strong> اجازهٔ تولید Claim مستقل و اصلی دارد.</li>
<li><strong>Secondary Claim Discovery:</strong> اجازهٔ تولید Claim دارد، اما Attribution و Authority ثانویه باید حفظ شود.</li>
<li><strong>Context Only:</strong> می‌تواند Context یا Supporting Evidence باشد، اما به‌صورت خودکار Claim مستقل دانش اصلی ایجاد نمی‌کند.</li>
<li><strong>Unresolved Hold:</strong> تا زمان حل Attribution اجازهٔ ورود به Registry پذیرفته‌شده را ندارد.</li>
</ul>

<p>
این تفکیک باعث می‌شود Claim Registry بزرگ و قابل استفاده برای کتاب باشد، بدون اینکه سؤال مجری، Context گفتگو یا Speaker نامشخص به‌اشتباه به دانش اصلی مؤلف تبدیل شود.
</p>

<h3>Source of Truthهای این Notebook</h3>

<p>
این مرحله باید به‌صورت Read-Only تأیید کند که:
</p>

<ul>
<li>Accepted Corpus همان Snapshot فریز‌شدهٔ قبلی است؛</li>
<li>Claim Discovery Population Index تمام Accepted Recordها را پوشش می‌دهد؛</li>
<li>هر <code>document_id</code> دقیقاً یک Role دارد؛</li>
<li>Population Index و Accepted Corpus از نظر Document ID کاملاً قابل Join هستند؛</li>
<li>Authority و Claim Discovery Roleهای قبلی بدون بازسازی مجدد قابل استفاده‌اند؛</li>
<li>Dedup و Structure Auditهای NB12h به‌عنوان Supporting Contract قابل دسترس هستند؛</li>
<li>Claim Registry جدید از هیچ Artifact قبلی با نام مشابه ادامه داده نمی‌شود.</li>
</ul>

<h3>چرا Authority را دوباره محاسبه نمی‌کنیم؟</h3>

<p>
Authority و Attribution قبلاً در Notebookهای قبلی با Evidence واقعی تعیین و Freeze شده‌اند.
این Notebook آن منطق را دوباره اختراع یا بازسازی نمی‌کند؛ فقط Artifact پذیرفته‌شده را مصرف و Validate می‌کند.
</p>

<h3>خروجی این مرحله</h3>

<p>
در این مرحله هنوز Artifact جدیدی ساخته نمی‌شود.
خروجی فقط یک Input Contract معتبر در حافظهٔ Notebook است که مشخص می‌کند:
</p>

<ul>
<li>چند Record در کل Corpus وجود دارد؛</li>
<li>چند Record اجازهٔ Claim Discovery دارند؛</li>
<li>چند Record فقط Context هستند؛</li>
<li>چند Record در Hold قرار دارند؛</li>
<li>Population هر Source Type چیست؛</li>
<li>و آیا Join میان Corpus و Population Index کامل است یا خیر.</li>
</ul>

<h3>معیار PASS</h3>

<ul>
<li>Accepted Corpus دقیقاً با Snapshot مرحلهٔ قبل یکسان باشد؛</li>

</p>

```python
# Code Cell 03

assert "project_root" in globals()
assert "accepted_corpus_path" in globals()
assert "accepted_corpus_profile" in globals()
assert "relevant_artifacts" in globals()
assert "existing_registry_artifacts" in globals()


def load_json(path):
    with path.open(
        "r",
        encoding="utf-8",
        errors="strict",
    ) as file:
        return json.load(file)


def load_jsonl(path):
    rows = []

    with path.open(
        "r",
        encoding="utf-8",
        errors="strict",
    ) as file:

        for line_number, line in enumerate(
            file,
            1,
        ):
            stripped = line.strip()

            if not stripped:
                continue

            row = json.loads(
                stripped
            )

            assert isinstance(
                row,
                dict,
            ), (
                "Non-object JSONL record at "
                f"{path}:{line_number}"
            )

            rows.append(row)

    return rows


# ---------------------------------------------------------
# Resolve prerequisite artifacts discovered by Cell 02.
# ---------------------------------------------------------

population_index_matches = (
    relevant_artifacts[
        "claim_discovery_population_index_v1.jsonl"
    ]
)

population_prep_matches = (
    relevant_artifacts[
        "claim_discovery_population_prep_v1.json"
    ]
)

dedup_topology_matches = (
    relevant_artifacts[
        "claim_discovery_dedup_topology_audit_v1.json"
    ]
)

chunk_independence_matches = (
    relevant_artifacts[
        "claim_discovery_chunk_independence_audit_v1.json"
    ]
)

structure_contract_matches = (
    relevant_artifacts[
        "claim_extraction_structure_contract_audit_v1.json"
    ]
)


required_singletons = {
    "population_index": (
        population_index_matches
    ),
    "population_prep": (
        population_prep_matches
    ),
    "dedup_topology": (
        dedup_topology_matches
    ),
    "chunk_independence": (
        chunk_independence_matches
    ),
    "structure_contract": (
        structure_contract_matches
    ),
}


singleton_failures = {
    name: len(paths)
    for name, paths in (
        required_singletons.items()
    )
    if len(paths) != 1
}


assert not singleton_failures, (
    "Required prerequisite artifact "
    "is not unique: "
    f"{singleton_failures}"
)


population_index_path = (
    population_index_matches[0]
)

population_prep_path = (
    population_prep_matches[0]
)

dedup_topology_path = (
    dedup_topology_matches[0]
)

chunk_independence_path = (
    chunk_independence_matches[0]
)

structure_contract_path = (
    structure_contract_matches[0]
)


# ---------------------------------------------------------
# Load frozen prerequisite artifacts.
# ---------------------------------------------------------

population_rows = load_jsonl(
    population_index_path
)

population_prep = load_json(
    population_prep_path
)

dedup_topology = load_json(
    dedup_topology_path
)

chunk_independence = load_json(
    chunk_independence_path
)

structure_contract = load_json(
    structure_contract_path
)


# ---------------------------------------------------------
# Independently re-read Accepted Corpus.
# ---------------------------------------------------------

accepted_document_ids = set()

accepted_source_type_counts = Counter()

accepted_duplicate_document_ids = []


with accepted_corpus_path.open(
    "r",
    encoding="utf-8",
    errors="strict",
) as file:

    for line_number, line in enumerate(
        file,
        1,
    ):
        stripped = line.strip()

        if not stripped:
            continue

        row = json.loads(
            stripped
        )

        document_id = row.get(
            "document_id"
        )

        source_type = row.get(
            "source_type"
        )

        assert isinstance(
            document_id,
            str,
        ) and document_id.strip(), (
            "Accepted Corpus contains "
            "a missing document_id at line "
            f"{line_number}."
        )

        if document_id in (
            accepted_document_ids
        ):
            accepted_duplicate_document_ids.append(
                document_id
            )

        accepted_document_ids.add(
            document_id
        )

        accepted_source_type_counts[
            source_type
        ] += 1


assert not (
    accepted_duplicate_document_ids
), (
    "Duplicate document_id detected "
    "in Accepted Corpus."
)


# ---------------------------------------------------------
# Validate Claim Discovery Population Index.
# ---------------------------------------------------------

population_document_ids = []

population_role_counts = Counter()

population_authority_counts = Counter()

population_source_type_counts = Counter()

unknown_roles = []


allowed_roles = {
    "PRIMARY_CLAIM_DISCOVERY",
    "SECONDARY_CLAIM_DISCOVERY_WITH_ATTRIBUTION",
    "CONTEXT_ONLY_NO_INDEPENDENT_SEED",
    "HOLD_UNRESOLVED_NO_INDEPENDENT_SEED",
}


for row in population_rows:

    document_id = row.get(
        "document_id"
    )

    role = row.get(
        "claim_discovery_role"
    )

    authority_class = row.get(
        "authority_class"
    )

    source_type = row.get(
        "source_type"
    )

    assert isinstance(
        document_id,
        str,
    ) and document_id.strip(), (
        "Population Index contains "
        "a missing document_id."
    )

    population_document_ids.append(
        document_id
    )

    population_role_counts[
        role
    ] += 1

    population_authority_counts[
        authority_class
    ] += 1

    population_source_type_counts[
        source_type
    ] += 1

    if role not in allowed_roles:
        unknown_roles.append(
            role
        )


population_document_id_set = set(
    population_document_ids
)


population_duplicate_count = (
    len(
        population_document_ids
    )
    - len(
        population_document_id_set
    )
)


missing_from_population_index = sorted(
    accepted_document_ids
    - population_document_id_set
)


extra_in_population_index = sorted(
    population_document_id_set
    - accepted_document_ids
)


# ---------------------------------------------------------
# Derive population classes.
# ---------------------------------------------------------

claim_discovery_eligible_count = (
    population_role_counts[
        "PRIMARY_CLAIM_DISCOVERY"
    ]
    +
    population_role_counts[
        "SECONDARY_CLAIM_DISCOVERY_WITH_ATTRIBUTION"
    ]
)


context_only_count = (
    population_role_counts[
        "CONTEXT_ONLY_NO_INDEPENDENT_SEED"
    ]
)


unresolved_hold_count = (
    population_role_counts[
        "HOLD_UNRESOLVED_NO_INDEPENDENT_SEED"
    ]
)


supporting_report_statuses = {
    "population_prep": (
        population_prep.get(
            "status",
            "NO_TOP_LEVEL_STATUS",
        )
    ),
    "dedup_topology": (
        dedup_topology.get(
            "status",
            "NO_TOP_LEVEL_STATUS",
        )
    ),
    "chunk_independence": (
        chunk_independence.get(
            "status",
            "NO_TOP_LEVEL_STATUS",
        )
    ),
    "structure_contract": (
        structure_contract.get(
            "status",
            "NO_TOP_LEVEL_STATUS",
        )
    ),
}


# ---------------------------------------------------------
# Input Contract validation.
#
# IMPORTANT:
# Existing Registry artifacts are NOT a failure here.
#
# This Cell validates immutable INPUTS.
# On Restart Kernel -> Run All, outputs created by later
# Cells are expected to already exist and are independently
# revalidated by their own no-overwrite/reuse logic.
# ---------------------------------------------------------

contract_failures = []


if (
    len(
        accepted_document_ids
    )
    != accepted_corpus_profile[
        "records"
    ]
):
    contract_failures.append(
        "Accepted Corpus count differs "
        "from Preflight snapshot."
    )


if (
    file_sha256(
        accepted_corpus_path
    )
    != accepted_corpus_profile[
        "sha256"
    ]
):
    contract_failures.append(
        "Accepted Corpus SHA-256 changed "
        "after Preflight."
    )


if (
    len(
        population_rows
    )
    != len(
        accepted_document_ids
    )
):
    contract_failures.append(
        "Population Index record count "
        "does not equal Accepted Corpus."
    )


if population_duplicate_count != 0:
    contract_failures.append(
        "Population Index contains "
        "duplicate document IDs."
    )


if missing_from_population_index:
    contract_failures.append(
        "Accepted Corpus contains documents "
        "missing from Population Index."
    )


if extra_in_population_index:
    contract_failures.append(
        "Population Index contains documents "
        "outside Accepted Corpus."
    )


if unknown_roles:
    contract_failures.append(
        "Unknown Claim Discovery role found."
    )


# ---------------------------------------------------------
# Freeze validated Input Contract in memory.
# ---------------------------------------------------------

claim_registry_input_contract = {
    "accepted_corpus_path": str(
        accepted_corpus_path.relative_to(
            project_root
        )
    ),

    "accepted_corpus_records": len(
        accepted_document_ids
    ),

    "accepted_corpus_sha256": (
        accepted_corpus_profile[
            "sha256"
        ]
    ),

    "population_index_path": str(
        population_index_path.relative_to(
            project_root
        )
    ),

    "population_index_records": len(
        population_rows
    ),

    "record_counts_by_source_type": dict(
        sorted(
            population_source_type_counts.items()
        )
    ),

    "authority_counts": dict(
        sorted(
            population_authority_counts.items()
        )
    ),

    "claim_discovery_role_counts": dict(
        sorted(
            population_role_counts.items()
        )
    ),

    "claim_discovery_eligible_records": (
        claim_discovery_eligible_count
    ),

    "context_only_records": (
        context_only_count
    ),

    "unresolved_hold_records": (
        unresolved_hold_count
    ),

    "supporting_report_statuses": (
        supporting_report_statuses
    ),

    "registry_population_policy": {
        "primary": (
            "INDEPENDENT_CLAIM_DISCOVERY_ALLOWED"
        ),

        "secondary": (
            "CLAIM_DISCOVERY_ALLOWED_WITH_"
            "SECONDARY_ATTRIBUTION"
        ),

        "context": (
            "CONTEXT_OR_SUPPORT_ONLY"
        ),

        "unresolved": (
            "HOLD_UNTIL_ATTRIBUTION_RESOLVED"
        ),
    },
}


# ---------------------------------------------------------
# Human-readable result.
# ---------------------------------------------------------

print(
    "=== NB12i CLAIM REGISTRY INPUT CONTRACT ==="
)

print()

print(
    "Accepted Corpus records:",
    len(
        accepted_document_ids
    ),
)

print(
    "Accepted Corpus SHA-256:",
    accepted_corpus_profile[
        "sha256"
    ],
)


print()

print(
    "Population Index records:",
    len(
        population_rows
    ),
)

print(
    "Duplicate document IDs:",
    population_duplicate_count,
)

print(
    "Missing from Population Index:",
    len(
        missing_from_population_index
    ),
)

print(
    "Extra in Population Index:",
    len(
        extra_in_population_index
    ),
)


print()

print(
    "Records by source type:",
    dict(
        sorted(
            population_source_type_counts.items()
        )
    ),
)


print()

print(
    "Authority classes:",
    dict(
        sorted(
            population_authority_counts.items()
        )
    ),
)


print()

print(
    "Claim Discovery roles:",
    dict(
        sorted(
            population_role_counts.items()
        )
    ),
)


print()

print(
    "Claim Discovery eligible records:",
    claim_discovery_eligible_count,
)

print(
    "Context-only records:",
    context_only_count,
)

print(
    "Unresolved hold records:",
    unresolved_hold_count,
)


print()

print(
    "Supporting report statuses:",
    supporting_report_statuses,
)


print()

print(
    "Existing registry artifacts:",
    len(
        existing_registry_artifacts
    ),
)

print(
    "Existing registry artifacts allowed on rerun:",
    True,
)


print()

print(
    "Contract failures:",
    len(
        contract_failures
    ),
)


for failure in contract_failures:
    print(
        "  -",
        failure,
    )


claim_registry_input_contract_status = (
    "PASS"
    if not contract_failures
    else "FAIL"
)


print()

print(
    "CODE CELL 03 GATE:",
    claim_registry_input_contract_status,
)
```

<li>Claim، Evidence و Relation از هم تفکیک شده باشند؛</li>
<li>Raw Claim ID و Final Claim ID با هم قاطی نشده باشند؛</li>
<li>Authority و Attribution در Schema حفظ شده باشند؛</li>
<li>Golden، Book و GraphRAG به‌عنوان Consumer تعریف شده باشند، نه Owner داده؛</li>
<li>Contract قبل از Publication در حافظه Validate شود؛</li>
<li>Artifact موجود Overwrite نشود؛</li>
<li>پس از نوشتن، فایل از Disk دوباره Parse و Hash شود؛</li>
<li>هیچ Claim Extraction یا API Call در این مرحله انجام نشود.</li>
</ul>

```python
# Code Cell 04

assert (
    claim_registry_input_contract_status
    == "PASS"
), (
    "Code Cell 03 must PASS before "
    "building the registry contract."
)


registry_artifact_root = (
    project_root
    / "artifacts"
    / "phase2"
    / "knowledge_registry"
)

registry_contract_path = (
    registry_artifact_root
    / "claim_registry_contract_v1.json"
)


claim_registry_contract = {
    "contract_version": (
        "claim_registry_contract_v1"
    ),

    "purpose": (
        "CENTRAL_KNOWLEDGE_ASSET_FOR_"
        "RAG_GOLDEN_BOOK_AND_GRAPH"
    ),

    "source_of_truth": {
        "accepted_corpus": {
            "path": (
                claim_registry_input_contract[
                    "accepted_corpus_path"
                ]
            ),
            "records": (
                claim_registry_input_contract[
                    "accepted_corpus_records"
                ]
            ),
            "sha256": (
                claim_registry_input_contract[
                    "accepted_corpus_sha256"
                ]
            ),
        },

        "claim_discovery_population": {
            "path": (
                claim_registry_input_contract[
                    "population_index_path"
                ]
            ),
            "records": (
                claim_registry_input_contract[
                    "population_index_records"
                ]
            ),
            "eligible_records": (
                claim_registry_input_contract[
                    "claim_discovery_eligible_records"
                ]
            ),
            "context_only_records": (
                claim_registry_input_contract[
                    "context_only_records"
                ]
            ),
            "unresolved_hold_records": (
                claim_registry_input_contract[
                    "unresolved_hold_records"
                ]
            ),
        },

        "authority_policy_source": (
            "FROZEN_PRIOR_PHASE_ARTIFACTS"
        ),

        "authority_recalculation_allowed": False,
    },

    "registry_population_policy": (
        claim_registry_input_contract[
            "registry_population_policy"
        ]
    ),

    "artifact_namespace": (
        "artifacts/phase2/knowledge_registry"
    ),

    "artifact_family": {
        "raw_claims": (
            "claims_raw_v1.jsonl"
        ),
        "validated_claims": (
            "claims_validated_v1.jsonl"
        ),
        "deduplicated_claims": (
            "claims_deduplicated_v1.jsonl"
        ),
        "evidence_links": (
            "claim_evidence_links_v1.jsonl"
        ),
        "claim_relations": (
            "claim_relations_v1.jsonl"
        ),
        "knowledge_graph": (
            "knowledge_graph_v1.json"
        ),
        "registry_report": (
            "claim_registry_report_v1.json"
        ),
    },

    "identity_policy": {
        "raw_claim_id": {
            "stage": (
                "EXTRACTION_AND_PRE_DEDUP"
            ),
            "required": True,
            "stable_within_extraction_contract": True,
            "purpose": (
                "TRACE_RAW_CLAIM_TO_EXTRACTION_"
                "UNIT_AND_EVIDENCE"
            ),
        },

        "claim_id": {
            "stage": (
                "POST_VALIDATION_AND_DEDUP"
            ),
            "required_before_registry_freeze": True,
            "assigned_during_raw_extraction": False,
            "immutable_after_registry_freeze": True,
        },

        "early_canonicalization_allowed": False,
    },

    "raw_claim_schema": {
        "required_fields": [
            "raw_claim_id",
            "claim_text_fa",
            "claim_type",
            "source_document_ids",
            "authority_class",
            "claim_discovery_role",
            "extraction_unit_id",
            "grounding_status",
            "atomicity_status",
            "validation_status",
        ],

        "optional_fields": [
            "topic_id",
            "subtopic_id",
            "notes",
        ],

        "field_rules": {
            "raw_claim_id": (
                "NONEMPTY_STABLE_STRING"
            ),
            "claim_text_fa": (
                "NONEMPTY_PERSIAN_CLAIM"
            ),
            "claim_type": (
                "CONTROLLED_ENUM_TO_BE_"
                "VALIDATED_BEFORE_EXTRACTION"
            ),
            "topic_id": (
                "MUST_REFERENCE_FROZEN_TAXONOMY_"
                "WHEN_PRESENT"
            ),
            "source_document_ids": (
                "NONEMPTY_LIST_OF_ACCEPTED_"
                "DOCUMENT_IDS"
            ),
            "authority_class": (
                "MUST_MATCH_POPULATION_INDEX"
            ),
            "claim_discovery_role": (
                "MUST_MATCH_POPULATION_INDEX"
            ),
            "extraction_unit_id": (
                "MUST_REFERENCE_EXTRACTION_UNIT"
            ),
        },
    },

    "evidence_link_schema": {
        "required_fields": [
            "evidence_id",
            "document_id",
            "source_id",
            "source_type",
            "authority_class",
            "claim_discovery_role",
            "evidence_quote_exact",
            "locator",
            "attribution_snapshot",
            "evidence_text_sha256",
        ],

        "claim_reference_rule": (
            "MUST_REFERENCE_RAW_CLAIM_ID_"
            "BEFORE_DEDUP_OR_CLAIM_ID_"
            "AFTER_DEDUP"
        ),

        "field_rules": {
            "document_id": (
                "MUST_EXIST_IN_ACCEPTED_CORPUS"
            ),
            "source_id": (
                "MUST_MATCH_ACCEPTED_DOCUMENT"
            ),
            "source_type": (
                "MUST_MATCH_ACCEPTED_DOCUMENT"
            ),
            "authority_class": (
                "MUST_MATCH_POPULATION_INDEX"
            ),
            "claim_discovery_role": (
                "MUST_MATCH_POPULATION_INDEX"
            ),
            "evidence_quote_exact": (
                "MUST_BE_VERBATIM_SOURCE_SPAN"
            ),
            "locator": (
                "MUST_PRESERVE_REAL_SOURCE_LOCATOR"
            ),
            "attribution_snapshot": (
                "MUST_PRESERVE_SOURCE_ATTRIBUTION"
            ),
        },
    },

    "dedup_relation_types": [
        "EXACT_DUPLICATE",
        "SEMANTIC_DUPLICATE",
        "SUBSUMES",
        "PARTIAL_OVERLAP",
    ],

    "knowledge_relation_types": [
        "RELATED_TO",
        "SUPPORTS",
        "CONTRADICTS",
        "PREREQUISITE_OF",
        "BROADER_THAN",
        "NARROWER_THAN",
    ],

    "relation_policy": {
        "automatic_semantic_merge_allowed": False,
        "automatic_subsumption_merge_allowed": False,
        "source_identity_must_be_preserved": True,
        "evidence_multiplicity_must_be_preserved": True,
        "relation_requires_validated_stage": True,
    },

    "consumer_contract": {
        "golden_dataset": {
            "role": (
                "REPRESENTATIVE_SELECTION_FROM_"
                "ACCEPTED_CLAIMS"
            ),
            "full_registry_required": False,
        },

        "book_production": {
            "role": (
                "CLAIM_AND_EVIDENCE_KNOWLEDGE_BASE"
            ),
            "may_consume_full_registry": True,
        },

        "graphrag": {
            "role": (
                "FUTURE_GRAPH_ASSISTED_RETRIEVAL"
            ),
            "runtime_dependency_now": False,
        },
    },

    "source_corpus_policy": {
        "claim_registry_replaces_source_corpus": False,
        "original_source_remains_final_evidence": True,
        "retrieval_chunks_remain_separate": True,
    },

    "build_policy": {
        "batch_based": True,
        "resume_safe": True,
        "cache_required_for_model_calls": True,
        "no_overwrite": True,
        "versioned_artifacts": True,
        "independent_disk_revalidation": True,
    },

    "current_state": {
        "claim_extraction_started": False,
        "raw_claims_created": False,
        "validated_claims_created": False,
        "deduplicated_claims_created": False,
        "knowledge_graph_created": False,
        "golden_dataset_v2_created": False,
    },
}


required_top_level_keys = {
    "contract_version",
    "purpose",
    "source_of_truth",
    "registry_population_policy",
    "artifact_namespace",
    "artifact_family",
    "identity_policy",
    "raw_claim_schema",
    "evidence_link_schema",
    "dedup_relation_types",
    "knowledge_relation_types",
    "relation_policy",
    "consumer_contract",
    "source_corpus_policy",
    "build_policy",
    "current_state",
}


contract_validation_failures = []


missing_top_level_keys = sorted(
    required_top_level_keys
    - set(
        claim_registry_contract
    )
)

if missing_top_level_keys:
    contract_validation_failures.append(
        "Missing top-level contract keys: "
        + ", ".join(
            missing_top_level_keys
        )
    )


if (
    claim_registry_contract[
        "source_of_truth"
    ][
        "accepted_corpus"
    ][
        "records"
    ]
    != 19456
):
    contract_validation_failures.append(
        "Accepted Corpus count differs "
        "from validated Input Contract."
    )


if (
    claim_registry_contract[
        "source_of_truth"
    ][
        "claim_discovery_population"
    ][
        "eligible_records"
    ]
    != 15158
):
    contract_validation_failures.append(
        "Eligible population differs "
        "from validated Input Contract."
    )


if (
    claim_registry_contract[
        "identity_policy"
    ][
        "early_canonicalization_allowed"
    ]
    is not False
):
    contract_validation_failures.append(
        "Early canonicalization must remain disabled."
    )


if (
    claim_registry_contract[
        "source_corpus_policy"
    ][
        "claim_registry_replaces_source_corpus"
    ]
    is not False
):
    contract_validation_failures.append(
        "Claim Registry must not replace Source Corpus."
    )


if (
    claim_registry_contract[
        "relation_policy"
    ][
        "automatic_semantic_merge_allowed"
    ]
    is not False
):
    contract_validation_failures.append(
        "Automatic semantic merge must remain disabled."
    )


raw_required_fields = set(
    claim_registry_contract[
        "raw_claim_schema"
    ][
        "required_fields"
    ]
)

required_raw_identity_fields = {
    "raw_claim_id",
    "claim_text_fa",
    "source_document_ids",
    "authority_class",
    "claim_discovery_role",
    "extraction_unit_id",
}

missing_raw_identity_fields = sorted(
    required_raw_identity_fields
    - raw_required_fields
)

if missing_raw_identity_fields:
    contract_validation_failures.append(
        "Raw Claim schema is missing "
        "required identity fields: "
        + ", ".join(
            missing_raw_identity_fields
        )
    )


evidence_required_fields = set(
    claim_registry_contract[
        "evidence_link_schema"
    ][
        "required_fields"
    ]
)

required_evidence_provenance_fields = {
    "evidence_id",
    "document_id",
    "source_id",
    "source_type",
    "evidence_quote_exact",
    "locator",
    "attribution_snapshot",
}

missing_evidence_fields = sorted(
    required_evidence_provenance_fields
    - evidence_required_fields
)

if missing_evidence_fields:
    contract_validation_failures.append(
        "Evidence schema is missing "
        "required provenance fields: "
        + ", ".join(
            missing_evidence_fields
        )
    )


assert not contract_validation_failures, (
    "Claim Registry Contract validation failed:\n- "
    + "\n- ".join(
        contract_validation_failures
    )
)


staged_contract_text = (
    json.dumps(
        claim_registry_contract,
        ensure_ascii=False,
        indent=2,
        sort_keys=True,
    )
    + "\n"
)

staged_contract_sha256 = hashlib.sha256(
    staged_contract_text.encode(
        "utf-8"
    )
).hexdigest()


registry_artifact_root.mkdir(
    parents=True,
    exist_ok=True,
)


if registry_contract_path.exists():

    existing_contract_text = (
        registry_contract_path.read_text(
            encoding="utf-8"
        )
    )

    assert (
        existing_contract_text
        == staged_contract_text
    ), (
        "Existing Claim Registry Contract differs "
        "from the validated staged contract. "
        "Refusing to overwrite."
    )

    contract_write_status = (
        "REUSED_EXISTING_IDENTICAL_CONTRACT"
    )

else:

    with registry_contract_path.open(
        "x",
        encoding="utf-8",
    ) as file:
        file.write(
            staged_contract_text
        )

    contract_write_status = (
        "CREATED_NEW_CONTRACT"
    )


with registry_contract_path.open(
    "r",
    encoding="utf-8",
    errors="strict",
) as file:
    disk_contract = json.load(
        file
    )


disk_contract_text = (
    registry_contract_path.read_text(
        encoding="utf-8"
    )
)

disk_contract_sha256 = hashlib.sha256(
    disk_contract_text.encode(
        "utf-8"
    )
).hexdigest()


assert (
    disk_contract
    == claim_registry_contract
), (
    "Disk re-validation failed: "
    "parsed contract differs from staged contract."
)

assert (
    disk_contract_sha256
    == staged_contract_sha256
), (
    "Disk re-validation failed: "
    "contract SHA-256 mismatch."
)


claim_registry_contract_status = "PASS"


print(
    "=== NB12i CLAIM REGISTRY CONTRACT ==="
)

print()
print(
    "Contract version:",
    claim_registry_contract[
        "contract_version"
    ],
)

print(
    "Artifact namespace:",
    claim_registry_contract[
        "artifact_namespace"
    ],
)

print()
print(
    "Accepted Corpus records:",
    claim_registry_contract[
        "source_of_truth"
    ][
        "accepted_corpus"
    ][
        "records"
    ],
)

print(
    "Claim Discovery eligible records:",
    claim_registry_contract[
        "source_of_truth"
    ][
        "claim_discovery_population"
    ][
        "eligible_records"
    ],
)

print()
print(
    "Artifact family:"
)

for (
    artifact_role,
    artifact_name,
) in (
    claim_registry_contract[
        "artifact_family"
    ].items()
):
    print(
        "  -",
        artifact_role,
        "->",
        artifact_name,
    )

print()
print(
    "Raw Claim required fields:",
    len(
        claim_registry_contract[
            "raw_claim_schema"
        ][
            "required_fields"
        ]
    ),
)

print(
    "Evidence required fields:",
    len(
        claim_registry_contract[
            "evidence_link_schema"
        ][
            "required_fields"
        ]
    ),
)

print(
    "Dedup relation types:",
    claim_registry_contract[
        "dedup_relation_types"
    ],
)

print(
    "Knowledge relation types:",
    claim_registry_contract[
        "knowledge_relation_types"
    ],
)

print()
print(
    "Early canonicalization allowed:",
    claim_registry_contract[
        "identity_policy"
    ][
        "early_canonicalization_allowed"
    ],
)

print(
    "Automatic semantic merge allowed:",
    claim_registry_contract[
        "relation_policy"
    ][
        "automatic_semantic_merge_allowed"
    ],
)

print(
    "Claim Registry replaces Source Corpus:",
    claim_registry_contract[
        "source_corpus_policy"
    ][
        "claim_registry_replaces_source_corpus"
    ],
)

print()
print(
    "Contract path:",
    registry_contract_path.relative_to(
        project_root
    ),
)

print(
    "Contract SHA-256:",
    disk_contract_sha256,
)

print(
    "Write status:",
    contract_write_status,
)

print()
print(
    "Claim extraction started:",
    False,
)

print(
    "Raw claims created:",
    False,
)

print(
    "Knowledge graph created:",
    False,
)

print()
print(
    "CODE CELL 04 GATE:",
    claim_registry_contract_status,
)
```

</div>

```python
# Code Cell 05

assert (
    claim_registry_contract_status
    == "PASS"
), (
    "Code Cell 04 must PASS before "
    "building extraction units."
)


unit_contract_path = (
    registry_artifact_root
    / "claim_extraction_unit_contract_v1.json"
)

unit_manifest_path = (
    registry_artifact_root
    / "claim_extraction_units_v1.jsonl"
)


ELIGIBLE_CLAIM_ROLES = {
    "PRIMARY_CLAIM_DISCOVERY",
    "SECONDARY_CLAIM_DISCOVERY_WITH_ATTRIBUTION",
}


def stable_json(value):
    return json.dumps(
        value,
        ensure_ascii=False,
        sort_keys=True,
        separators=(",", ":"),
    )


def stable_unit_id(
    source_type,
    source_id,
    document_ids,
):
    material = stable_json({
        "source_type": source_type,
        "source_id": source_id,
        "document_ids": document_ids,
    })

    digest = hashlib.sha256(
        material.encode("utf-8")
    ).hexdigest()[:24]

    return (
        f"claim-unit:{source_type}:{digest}"
    )


def build_book_unit(
    source_id,
    group,
):
    document_ids = [
        item[
            "index_row"
        ][
            "document_id"
        ]
        for item in group
    ]

    first_item = group[0]
    last_item = group[-1]

    return {
        "extraction_unit_id": (
            stable_unit_id(
                "book",
                source_id,
                document_ids,
            )
        ),
        "unit_version": (
            "claim_extraction_unit_v1"
        ),
        "unit_level": (
            "STRUCTURAL_EXTRACTION_REGION"
        ),
        "source_type": (
            "book"
        ),
        "source_id": (
            source_id
        ),
        "authority_class": (
            first_item[
                "index_row"
            ][
                "authority_class"
            ]
        ),
        "claim_discovery_role": (
            first_item[
                "index_row"
            ][
                "claim_discovery_role"
            ]
        ),
        "document_ids": (
            document_ids
        ),
        "document_count": len(
            document_ids
        ),
        "structure": {
            "policy": (
                "BOOK_CONTIGUOUS_SAME_HEADING"
            ),
            "heading_path": (
                first_item[
                    "heading_path"
                ]
            ),
            "paragraph_start": (
                first_item[
                    "paragraph_number"
                ]
            ),
            "paragraph_end": (
                last_item[
                    "paragraph_number"
                ]
            ),
        },
        "locator_range": {
            "first": (
                first_item[
                    "accepted_record"
                ].get(
                    "locator"
                )
            ),
            "last": (
                last_item[
                    "accepted_record"
                ].get(
                    "locator"
                )
            ),
        },
        "prompt_windowing_status": (
            "NOT_APPLIED_YET"
        ),
    }


def build_transcript_unit(
    source_id,
    group,
):
    document_ids = [
        item[
            "index_row"
        ][
            "document_id"
        ]
        for item in group
    ]

    first_item = group[0]
    last_item = group[-1]

    first_locator = (
        first_item[
            "accepted_record"
        ].get(
            "locator"
        )
        or {}
    )

    last_locator = (
        last_item[
            "accepted_record"
        ].get(
            "locator"
        )
        or {}
    )

    return {
        "extraction_unit_id": (
            stable_unit_id(
                "transcript",
                source_id,
                document_ids,
            )
        ),
        "unit_version": (
            "claim_extraction_unit_v1"
        ),
        "unit_level": (
            "STRUCTURAL_EXTRACTION_REGION"
        ),
        "source_type": (
            "transcript"
        ),
        "source_id": (
            source_id
        ),
        "authority_class": (
            first_item[
                "index_row"
            ][
                "authority_class"
            ]
        ),
        "claim_discovery_role": (
            first_item[
                "index_row"
            ][
                "claim_discovery_role"
            ]
        ),
        "document_ids": (
            document_ids
        ),
        "document_count": len(
            document_ids
        ),
        "structure": {
            "policy": (
                "TRANSCRIPT_CONTIGUOUS_"
                "AUTHORITY_HOMOGENEOUS_RUN"
            ),
            "utterance_start": (
                first_item[
                    "utterance_index"
                ]
            ),
            "utterance_end": (
                last_item[
                    "utterance_index"
                ]
            ),
        },
        "locator_range": {
            "program_id": (
                first_locator.get(
                    "program_id"
                )
            ),
            "episode_number": (
                first_locator.get(
                    "episode_number"
                )
            ),
            "source_line_start": (
                first_locator.get(
                    "source_line_start"
                )
            ),
            "source_line_end": (
                last_locator.get(
                    "source_line_end"
                )
            ),
        },
        "prompt_windowing_status": (
            "NOT_APPLIED_YET"
        ),
    }


accepted_records_by_id = {}

with accepted_corpus_path.open(
    "r",
    encoding="utf-8",
    errors="strict",
) as file:

    for line_number, line in enumerate(
        file,
        1,
    ):
        stripped = line.strip()

        if not stripped:
            continue

        record = json.loads(
            stripped
        )

        document_id = record.get(
            "document_id"
        )

        assert isinstance(
            document_id,
            str,
        ) and document_id.strip(), (
            "Accepted Corpus contains "
            "a missing document_id at line "
            f"{line_number}."
        )

        assert (
            document_id
            not in accepted_records_by_id
        ), (
            "Duplicate document_id in "
            "Accepted Corpus: "
            f"{document_id}"
        )

        accepted_records_by_id[
            document_id
        ] = record


population_by_id = {
    row["document_id"]: row
    for row in population_rows
}

assert (
    len(population_by_id)
    == 19456
), (
    "Population Index must contain "
    "19,456 unique document IDs."
)


eligible_rows = [
    row
    for row in population_rows
    if row[
        "claim_discovery_role"
    ] in ELIGIBLE_CLAIM_ROLES
]

assert (
    len(eligible_rows)
    == 15158
), (
    "Eligible population differs from "
    "the validated Input Contract."
)


for row in eligible_rows:

    document_id = row[
        "document_id"
    ]

    assert (
        document_id
        in accepted_records_by_id
    ), (
        "Eligible document is missing "
        "from Accepted Corpus: "
        f"{document_id}"
    )

    accepted_record = (
        accepted_records_by_id[
            document_id
        ]
    )

    assert (
        accepted_record.get(
            "source_id"
        )
        == row.get(
            "source_id"
        )
    ), (
        "source_id mismatch between "
        "Accepted Corpus and Population Index: "
        f"{document_id}"
    )

    assert (
        accepted_record.get(
            "source_type"
        )
        == row.get(
            "source_type"
        )
    ), (
        "source_type mismatch between "
        "Accepted Corpus and Population Index: "
        f"{document_id}"
    )


units = []


website_rows = sorted(
    [
        row
        for row in eligible_rows
        if row[
            "source_type"
        ] == "website"
    ],
    key=lambda row: (
        row[
            "source_id"
        ],
        row[
            "document_id"
        ],
    ),
)


for row in website_rows:

    document_id = row[
        "document_id"
    ]

    accepted_record = (
        accepted_records_by_id[
            document_id
        ]
    )

    unit = {
        "extraction_unit_id": (
            stable_unit_id(
                "website",
                row[
                    "source_id"
                ],
                [
                    document_id
                ],
            )
        ),
        "unit_version": (
            "claim_extraction_unit_v1"
        ),
        "unit_level": (
            "STRUCTURAL_EXTRACTION_REGION"
        ),
        "source_type": (
            "website"
        ),
        "source_id": (
            row[
                "source_id"
            ]
        ),
        "authority_class": (
            row[
                "authority_class"
            ]
        ),
        "claim_discovery_role": (
            row[
                "claim_discovery_role"
            ]
        ),
        "document_ids": [
            document_id
        ],
        "document_count": 1,
        "structure": {
            "policy": (
                "WEBSITE_DOCUMENT_ATOMIC"
            ),
        },
        "locator_range": (
            accepted_record.get(
                "locator"
            )
        ),
        "prompt_windowing_status": (
            "NOT_APPLIED_YET"
        ),
    }

    units.append(unit)


book_rows_by_source = defaultdict(
    list
)

for row in eligible_rows:

    if row[
        "source_type"
    ] != "book":
        continue

    accepted_record = (
        accepted_records_by_id[
            row[
                "document_id"
            ]
        ]
    )

    locator = (
        accepted_record.get(
            "locator"
        )
        or {}
    )

    paragraph_number = (
        locator.get(
            "paragraph_number"
        )
    )

    heading_path = (
        locator.get(
            "heading_path"
        )
    )

    if heading_path is None:
        heading_path = []

    book_rows_by_source[
        row[
            "source_id"
        ]
    ].append({
        "index_row": row,
        "paragraph_number": (
            paragraph_number
        ),
        "heading_path": (
            heading_path
        ),
        "accepted_record": (
            accepted_record
        ),
    })


for (
    source_id,
    source_rows,
) in sorted(
    book_rows_by_source.items()
):

    numbered_rows = [
        item
        for item in source_rows
        if isinstance(
            item[
                "paragraph_number"
            ],
            int,
        )
    ]

    unnumbered_rows = [
        item
        for item in source_rows
        if not isinstance(
            item[
                "paragraph_number"
            ],
            int,
        )
    ]

    numbered_rows.sort(
        key=lambda item: (
            item[
                "paragraph_number"
            ],
            item[
                "index_row"
            ][
                "document_id"
            ],
        )
    )

    current_group = []

    previous_paragraph_number = None
    previous_heading_key = None
    previous_authority = None
    previous_role = None

    for item in numbered_rows:

        row = item[
            "index_row"
        ]

        paragraph_number = (
            item[
                "paragraph_number"
            ]
        )

        heading_key = stable_json(
            item[
                "heading_path"
            ]
        )

        authority = row[
            "authority_class"
        ]

        role = row[
            "claim_discovery_role"
        ]

        starts_new_group = (
            current_group
            and (
                paragraph_number
                != previous_paragraph_number + 1
                or heading_key
                != previous_heading_key
                or authority
                != previous_authority
                or role
                != previous_role
            )
        )

        if starts_new_group:
            units.append(
                build_book_unit(
                    source_id,
                    current_group,
                )
            )
            current_group = []

        current_group.append(
            item
        )

        previous_paragraph_number = (
            paragraph_number
        )

        previous_heading_key = (
            heading_key
        )

        previous_authority = (
            authority
        )

        previous_role = (
            role
        )

    if current_group:
        units.append(
            build_book_unit(
                source_id,
                current_group,
            )
        )


    for item in sorted(
        unnumbered_rows,
        key=lambda value: (
            value[
                "index_row"
            ][
                "document_id"
            ]
        ),
    ):

        row = item[
            "index_row"
        ]

        document_id = row[
            "document_id"
        ]

        units.append({
            "extraction_unit_id": (
                stable_unit_id(
                    "book",
                    source_id,
                    [
                        document_id
                    ],
                )
            ),
            "unit_version": (
                "claim_extraction_unit_v1"
            ),
            "unit_level": (
                "STRUCTURAL_EXTRACTION_REGION"
            ),
            "source_type": (
                "book"
            ),
            "source_id": (
                source_id
            ),
            "authority_class": (
                row[
                    "authority_class"
                ]
            ),
            "claim_discovery_role": (
                row[
                    "claim_discovery_role"
                ]
            ),
            "document_ids": [
                document_id
            ],
            "document_count": 1,
            "structure": {
                "policy": (
                    "BOOK_UNNUMBERED_SINGLETON"
                ),
                "heading_path": (
                    item[
                        "heading_path"
                    ]
                ),
                "paragraph_start": None,
                "paragraph_end": None,
            },
            "locator_range": (
                item[
                    "accepted_record"
                ].get(
                    "locator"
                )
            ),
            "prompt_windowing_status": (
                "NOT_APPLIED_YET"
            ),
        })


transcript_rows_by_source = defaultdict(
    list
)

for row in eligible_rows:

    if row[
        "source_type"
    ] != "transcript":
        continue

    accepted_record = (
        accepted_records_by_id[
            row[
                "document_id"
            ]
        ]
    )

    source_metadata = (
        accepted_record.get(
            "source_metadata"
        )
        or {}
    )

    utterance_index = (
        source_metadata.get(
            "utterance_index"
        )
    )

    assert isinstance(
        utterance_index,
        int,
    ), (
        "Eligible Transcript record is "
        "missing integer utterance_index: "
        f"{row['document_id']}"
    )

    transcript_rows_by_source[
        row[
            "source_id"
        ]
    ].append({
        "index_row": row,
        "utterance_index": (
            utterance_index
        ),
        "accepted_record": (
            accepted_record
        ),
    })


for (
    source_id,
    source_rows,
) in sorted(
    transcript_rows_by_source.items()
):

    source_rows.sort(
        key=lambda item: (
            item[
                "utterance_index"
            ],
            item[
                "index_row"
            ][
                "document_id"
            ],
        )
    )

    current_group = []

    previous_utterance_index = None
    previous_authority = None
    previous_role = None

    for item in source_rows:

        row = item[
            "index_row"
        ]

        utterance_index = (
            item[
                "utterance_index"
            ]
        )

        authority = row[
            "authority_class"
        ]

        role = row[
            "claim_discovery_role"
        ]

        starts_new_group = (
            current_group
            and (
                utterance_index
                != previous_utterance_index + 1
                or authority
                != previous_authority
                or role
                != previous_role
            )
        )

        if starts_new_group:
            units.append(
                build_transcript_unit(
                    source_id,
                    current_group,
                )
            )
            current_group = []

        current_group.append(
            item
        )

        previous_utterance_index = (
            utterance_index
        )

        previous_authority = (
            authority
        )

        previous_role = (
            role
        )

    if current_group:
        units.append(
            build_transcript_unit(
                source_id,
                current_group,
            )
        )


units = sorted(
    units,
    key=lambda unit: (
        unit[
            "source_type"
        ],
        unit[
            "source_id"
        ],
        unit[
            "extraction_unit_id"
        ],
    ),
)


unit_ids = [
    unit[
        "extraction_unit_id"
    ]
    for unit in units
]

assert (
    len(unit_ids)
    == len(set(unit_ids))
), (
    "Duplicate extraction_unit_id detected."
)


covered_document_ids = []

mixed_source_units = 0
mixed_authority_units = 0
mixed_role_units = 0


for unit in units:

    unit_document_ids = (
        unit[
            "document_ids"
        ]
    )

    assert unit_document_ids, (
        "Extraction Unit has no document IDs."
    )

    source_ids = set()
    authority_classes = set()
    roles = set()

    for document_id in (
        unit_document_ids
    ):

        assert (
            document_id
            in population_by_id
        ), (
            "Extraction Unit references "
            "unknown document_id: "
            f"{document_id}"
        )

        population_row = (
            population_by_id[
                document_id
            ]
        )

        assert (
            population_row[
                "claim_discovery_role"
            ]
            in ELIGIBLE_CLAIM_ROLES
        ), (
            "Non-eligible document entered "
            "Claim Extraction Unit: "
            f"{document_id}"
        )

        source_ids.add(
            population_row[
                "source_id"
            ]
        )

        authority_classes.add(
            population_row[
                "authority_class"
            ]
        )

        roles.add(
            population_row[
                "claim_discovery_role"
            ]
        )

        covered_document_ids.append(
            document_id
        )

    if len(source_ids) != 1:
        mixed_source_units += 1

    if len(authority_classes) != 1:
        mixed_authority_units += 1

    if len(roles) != 1:
        mixed_role_units += 1


covered_document_id_set = set(
    covered_document_ids
)

eligible_document_id_set = {
    row[
        "document_id"
    ]
    for row in eligible_rows
}


duplicate_coverage_count = (
    len(covered_document_ids)
    - len(
        covered_document_id_set
    )
)

missing_eligible_documents = sorted(
    eligible_document_id_set
    - covered_document_id_set
)

extra_covered_documents = sorted(
    covered_document_id_set
    - eligible_document_id_set
)


assert duplicate_coverage_count == 0, (
    "At least one eligible document appears "
    "in more than one Extraction Unit."
)

assert not missing_eligible_documents, (
    "Eligible documents are missing from "
    "Extraction Unit coverage."
)

assert not extra_covered_documents, (
    "Extraction Units contain documents "
    "outside the eligible population."
)

assert mixed_source_units == 0, (
    "At least one Extraction Unit crosses "
    "a Source boundary."
)

assert mixed_authority_units == 0, (
    "At least one Extraction Unit mixes "
    "Authority classes."
)

assert mixed_role_units == 0, (
    "At least one Extraction Unit mixes "
    "Claim Discovery roles."
)


unit_counts_by_source = Counter(
    unit[
        "source_type"
    ]
    for unit in units
)

document_counts_by_source = Counter()

unit_size_distribution_by_source = defaultdict(
    Counter
)

for unit in units:

    source_type = unit[
        "source_type"
    ]

    document_counts_by_source[
        source_type
    ] += unit[
        "document_count"
    ]

    unit_size_distribution_by_source[
        source_type
    ][
        unit[
            "document_count"
        ]
    ] += 1


unit_contract = {
    "contract_version": (
        "claim_extraction_unit_contract_v1"
    ),

    "source_contract": (
        "claim_registry_contract_v1"
    ),

    "purpose": (
        "STRUCTURAL_CLAIM_EXTRACTION_REGIONS"
    ),

    "eligible_population_records": (
        len(
            eligible_document_id_set
        )
    ),

    "policies": {
        "website": (
            "ONE_ELIGIBLE_DOCUMENT_PER_REGION"
        ),
        "book": (
            "CONTIGUOUS_SAME_SOURCE_SAME_HEADING_"
            "SAME_AUTHORITY_SAME_ROLE"
        ),
        "book_unnumbered": (
            "SINGLETON"
        ),
        "transcript": (
            "CONTIGUOUS_SAME_SOURCE_SAME_AUTHORITY_"
            "SAME_ROLE_WITH_GAPS_BREAKING_RUN"
        ),
    },

    "retrieval_chunk_dependency": False,

    "prompt_windowing_applied": False,

    "exact_duplicate_sharing_applied": False,

    "context_only_claim_generation_allowed": False,

    "unresolved_claim_generation_allowed": False,

    "source_boundary_crossing_allowed": False,

    "mixed_authority_unit_allowed": False,

    "mixed_role_unit_allowed": False,

    "manifest": (
        "claim_extraction_units_v1.jsonl"
    ),
}


unit_contract_text = (
    json.dumps(
        unit_contract,
        ensure_ascii=False,
        indent=2,
        sort_keys=True,
    )
    + "\n"
)

unit_manifest_text = "".join(
    json.dumps(
        unit,
        ensure_ascii=False,
        sort_keys=True,
    )
    + "\n"
    for unit in units
)


unit_contract_sha256 = hashlib.sha256(
    unit_contract_text.encode(
        "utf-8"
    )
).hexdigest()

unit_manifest_sha256 = hashlib.sha256(
    unit_manifest_text.encode(
        "utf-8"
    )
).hexdigest()


if unit_contract_path.exists():

    existing_text = (
        unit_contract_path.read_text(
            encoding="utf-8"
        )
    )

    assert (
        existing_text
        == unit_contract_text
    ), (
        "Existing extraction-unit contract "
        "differs from the current validated "
        "contract. Refusing overwrite."
    )

    unit_contract_write_status = (
        "REUSED_EXISTING_IDENTICAL_CONTRACT"
    )

else:

    with unit_contract_path.open(
        "x",
        encoding="utf-8",
    ) as file:
        file.write(
            unit_contract_text
        )

    unit_contract_write_status = (
        "CREATED_NEW_CONTRACT"
    )


if unit_manifest_path.exists():

    existing_text = (
        unit_manifest_path.read_text(
            encoding="utf-8"
        )
    )

    assert (
        existing_text
        == unit_manifest_text
    ), (
        "Existing extraction-unit manifest "
        "differs from the current validated "
        "manifest. Refusing overwrite."
    )

    unit_manifest_write_status = (
        "REUSED_EXISTING_IDENTICAL_MANIFEST"
    )

else:

    with unit_manifest_path.open(
        "x",
        encoding="utf-8",
    ) as file:
        file.write(
            unit_manifest_text
        )

    unit_manifest_write_status = (
        "CREATED_NEW_MANIFEST"
    )


disk_unit_contract = load_json(
    unit_contract_path
)

disk_units = load_jsonl(
    unit_manifest_path
)


assert (
    disk_unit_contract
    == unit_contract
), (
    "Disk extraction-unit contract "
    "differs from staged contract."
)

assert (
    disk_units
    == units
), (
    "Disk extraction-unit manifest "
    "differs from staged manifest."
)


disk_unit_contract_sha256 = file_sha256(
    unit_contract_path
)

disk_unit_manifest_sha256 = file_sha256(
    unit_manifest_path
)


assert (
    disk_unit_contract_sha256
    == unit_contract_sha256
), (
    "Extraction-unit contract SHA-256 "
    "failed disk re-validation."
)

assert (
    disk_unit_manifest_sha256
    == unit_manifest_sha256
), (
    "Extraction-unit manifest SHA-256 "
    "failed disk re-validation."
)


claim_extraction_unit_status = "PASS"


print(
    "=== NB12i CLAIM EXTRACTION UNITS ==="
)

print()
print(
    "Eligible records:",
    len(
        eligible_document_id_set
    ),
)

print(
    "Extraction units:",
    len(
        units
    ),
)

print()
print(
    "Units by source type:",
    dict(
        sorted(
            unit_counts_by_source.items()
        )
    ),
)

print(
    "Documents covered by source type:",
    dict(
        sorted(
            document_counts_by_source.items()
        )
    ),
)

print()
print(
    "Duplicate document coverage:",
    duplicate_coverage_count,
)

print(
    "Missing eligible documents:",
    len(
        missing_eligible_documents
    ),
)

print(
    "Extra covered documents:",
    len(
        extra_covered_documents
    ),
)

print()
print(
    "Mixed-source units:",
    mixed_source_units,
)

print(
    "Mixed-authority units:",
    mixed_authority_units,
)

print(
    "Mixed-role units:",
    mixed_role_units,
)


print()
print(
    "Unit-size distributions:"
)

for source_type in sorted(
    unit_size_distribution_by_source
):

    distribution = dict(
        sorted(
            unit_size_distribution_by_source[
                source_type
            ].items()
        )
    )

    print(
        "  -",
        source_type,
        distribution,
    )


print()
print(
    "Prompt windowing applied:",
    False,
)

print(
    "Exact duplicate sharing applied:",
    False,
)

print(
    "Claim extraction started:",
    False,
)

print()
print(
    "Unit contract:",
    unit_contract_path.relative_to(
        project_root
    ),
)

print(
    "Unit contract SHA-256:",
    disk_unit_contract_sha256,
)

print(
    "Unit contract write status:",
    unit_contract_write_status,
)

print()
print(
    "Unit manifest:",
    unit_manifest_path.relative_to(
        project_root
    ),
)

print(
    "Unit manifest SHA-256:",
    disk_unit_manifest_sha256,
)

print(
    "Unit manifest write status:",
    unit_manifest_write_status,
)

print()
print(
    "CODE CELL 05 GATE:",
    claim_extraction_unit_status,
)
```

<h2>تثبیت Canonical Text Field و اندازه‌گیری واقعی Extraction Unitها</h2>

<p>
Extraction Unitهای ساختاری با پوشش کامل Population مجاز ساخته شدند.
اما Distribution مرحلهٔ قبل نشان داد اندازهٔ Structural Regionها بسیار متفاوت است و بعضی Book و Transcript Regionها شامل صدها Processing Record هستند.
</p>

<p>
بنابراین این Regionها هنوز مستقیماً برای LLM Prompt مناسب فرض نمی‌شوند.
قبل از هر Windowing باید دو چیز از خود دادهٔ واقعی مشخص شود:
</p>

<ul>
<li>متن اصلی Claim Extraction دقیقاً در کدام Field از Accepted Corpus قرار دارد؛</li>
<li>اندازهٔ واقعی Unitها از نظر حجم متن چقدر است.</li>
</ul>

<h3>اصل عدم حدس</h3>

<p>
این Notebook از روی حافظه یا Convention فرض نمی‌کند که Field متن حتماً <code>text</code> است.
ابتدا وجود و پوشش واقعی آن در تمام ۱۵٬۱۵۸ Record مجاز بررسی می‌شود.
</p>

<p>
اگر <code>text</code> در همهٔ Recordهای مجاز یک String غیرخالی باشد، همان Field به‌عنوان Canonical Extraction Text پذیرفته می‌شود.
اگر حتی بخشی از Population این Contract را نقض کند، مرحله FAIL خواهد شد و Candidate Fieldهای واقعی Corpus گزارش می‌شوند تا همان مرحله Repair شود.
</p>

<h3>اندازه‌گیری Unitها</h3>

<p>
پس از تثبیت Text Field، حجم Raw Text هر Extraction Unit اندازه‌گیری می‌شود.
در این مرحله هنوز هیچ Character Limit، Token Limit یا Prompt Window Size انتخاب نمی‌شود.
</p>

<p>
برای هر Source Type توزیع زیر گزارش خواهد شد:
</p>

<ul>
<li>تعداد Unit؛</li>
<li>مجموع Characterها؛</li>
<li>Minimum؛</li>
<li>Median؛</li>
<li>P75؛</li>
<li>P90؛</li>
<li>P95؛</li>
<li>P99؛</li>
<li>Maximum.</li>
</ul>

<p>
همچنین بزرگ‌ترین Unitها بدون Dump کردن متن Source گزارش می‌شوند تا مشخص شود Windowing مرحلهٔ بعد باید با چه دامنه‌ای از اندازه‌ها برخورد کند.
</p>

<h3>چرا هنوز Window Size تعیین نمی‌شود؟</h3>

<p>
Window Size باید از ترکیب دو Evidence تعیین شود:
</p>

<ol>
<li>Distribution واقعی متن که در این مرحله اندازه‌گیری می‌شود؛</li>
<li>Context Contract مدل واقعی Claim Extraction که در مرحلهٔ بعد از تنظیمات موجود پروژه Discovery و Validate می‌شود.</li>
</ol>

<p>
بنابراین هیچ مقدار دلخواه مانند ۴۰۰۰، ۶۰۰۰ یا ۸۰۰۰ Character در این مرحله وارد Pipeline نمی‌شود.
</p>

<h3>Artifact این مرحله</h3>

<p>
در صورت PASS، Profile زیر ثبت می‌شود:
</p>

<pre><code>artifacts/phase2/knowledge_registry/claim_extraction_text_profile_v1.json
</code></pre>

<p>
این Artifact فقط Canonical Text Contract و Size Distribution را ثبت می‌کند و هنوز Prompt Window یا Claim تولید نمی‌کند.
</p>

<h3>معیار PASS</h3>

<ul>
<li>هر ۱۵٬۱۵۸ Record مجاز دارای Canonical Text غیرخالی باشد؛</li>
<li>هیچ Document خارج از Extraction Unit Manifest وارد Profile نشود؛</li>
<li>تمام ۲٬۱۲۹ Extraction Unit قابل Materialize شدن باشند؛</li>
<li>هیچ Unit با متن خالی وجود نداشته باشد؛</li>
<li>Countهای Source Type با مرحلهٔ قبل سازگار باشند؛</li>
<li>هیچ Threshold دلخواه برای Windowing ساخته نشود؛</li>
<li>Profile با Policy عدم Overwrite منتشر شود؛</li>
<li>Artifact پس از نوشتن از Disk دوباره Parse و Hash شود؛</li>
<li>هیچ API یا LLM Call انجام نشود.</li>
</ul>

```python
# Code Cell 06

assert (
    claim_extraction_unit_status
    == "PASS"
), (
    "Code Cell 05 must PASS before "
    "profiling extraction text."
)


text_profile_path = (
    registry_artifact_root
    / "claim_extraction_text_profile_v1.json"
)


eligible_document_ids = {
    row["document_id"]
    for row in eligible_rows
}


missing_text_document_ids = []
empty_text_document_ids = []

top_level_string_field_coverage = Counter()
top_level_string_field_chars = Counter()


for document_id in sorted(
    eligible_document_ids
):

    record = (
        accepted_records_by_id[
            document_id
        ]
    )

    for key, value in record.items():

        if (
            isinstance(value, str)
            and value.strip()
        ):
            top_level_string_field_coverage[
                key
            ] += 1

            top_level_string_field_chars[
                key
            ] += len(value)

    if "text" not in record:
        missing_text_document_ids.append(
            document_id
        )

    else:

        value = record["text"]

        if (
            not isinstance(value, str)
            or not value.strip()
        ):
            empty_text_document_ids.append(
                document_id
            )


canonical_text_field = None

text_contract_failures = []


if missing_text_document_ids:
    text_contract_failures.append(
        "Eligible records are missing "
        "the top-level 'text' field."
    )


if empty_text_document_ids:
    text_contract_failures.append(
        "Eligible records contain an empty "
        "or non-string top-level 'text' field."
    )


if not text_contract_failures:
    canonical_text_field = "text"


candidate_string_fields = sorted(
    (
        {
            "field": field,
            "coverage": (
                top_level_string_field_coverage[
                    field
                ]
            ),
            "total_chars": (
                top_level_string_field_chars[
                    field
                ]
            ),
        }
        for field in (
            top_level_string_field_coverage
        )
    ),
    key=lambda item: (
        -item["coverage"],
        -item["total_chars"],
        item["field"],
    ),
)


def profile_quantile(
    sorted_values,
    fraction,
):
    assert sorted_values

    index = round(
        (
            len(sorted_values) - 1
        )
        * fraction
    )

    return sorted_values[
        index
    ]


unit_text_profiles = []


if canonical_text_field is not None:

    for unit in units:

        document_ids = (
            unit[
                "document_ids"
            ]
        )

        raw_text_chars = 0

        for document_id in (
            document_ids
        ):

            record = (
                accepted_records_by_id[
                    document_id
                ]
            )

            text = record[
                canonical_text_field
            ]

            assert isinstance(
                text,
                str,
            ) and text.strip(), (
                "Canonical text contract "
                "was violated during Unit "
                "materialization: "
                f"{document_id}"
            )

            raw_text_chars += len(
                text
            )

        assert raw_text_chars > 0, (
            "Extraction Unit materialized "
            "to zero text characters: "
            f"{unit['extraction_unit_id']}"
        )

        unit_text_profiles.append({
            "extraction_unit_id": (
                unit[
                    "extraction_unit_id"
                ]
            ),
            "source_type": (
                unit[
                    "source_type"
                ]
            ),
            "document_count": (
                unit[
                    "document_count"
                ]
            ),
            "raw_text_chars": (
                raw_text_chars
            ),
        })


source_text_profiles = {}


if canonical_text_field is not None:

    for source_type in sorted(
        unit_counts_by_source
    ):

        source_units = [
            item
            for item in unit_text_profiles
            if item[
                "source_type"
            ] == source_type
        ]

        char_values = sorted(
            item[
                "raw_text_chars"
            ]
            for item in source_units
        )

        source_text_profiles[
            source_type
        ] = {
            "units": len(
                source_units
            ),
            "total_raw_text_chars": sum(
                char_values
            ),
            "min_raw_text_chars": (
                char_values[0]
            ),
            "p50_raw_text_chars": (
                profile_quantile(
                    char_values,
                    0.50,
                )
            ),
            "p75_raw_text_chars": (
                profile_quantile(
                    char_values,
                    0.75,
                )
            ),
            "p90_raw_text_chars": (
                profile_quantile(
                    char_values,
                    0.90,
                )
            ),
            "p95_raw_text_chars": (
                profile_quantile(
                    char_values,
                    0.95,
                )
            ),
            "p99_raw_text_chars": (
                profile_quantile(
                    char_values,
                    0.99,
                )
            ),
            "max_raw_text_chars": (
                char_values[-1]
            ),
        }


largest_units = []


if canonical_text_field is not None:

    largest_units = sorted(
        unit_text_profiles,
        key=lambda item: (
            -item[
                "raw_text_chars"
            ],
            item[
                "extraction_unit_id"
            ],
        ),
    )[:10]


expected_unit_count = len(
    units
)

materialized_unit_count = len(
    unit_text_profiles
)


if (
    canonical_text_field is not None
    and materialized_unit_count
    != expected_unit_count
):
    text_contract_failures.append(
        "Not all Extraction Units were "
        "materialized for text profiling."
    )


if canonical_text_field is not None:

    profiled_source_counts = Counter(
        item[
            "source_type"
        ]
        for item in (
            unit_text_profiles
        )
    )

    if (
        profiled_source_counts
        != unit_counts_by_source
    ):
        text_contract_failures.append(
            "Source-type Unit counts changed "
            "during text profiling."
        )


text_profile = None
text_profile_write_status = (
    "NOT_WRITTEN"
)
disk_text_profile_sha256 = None


if not text_contract_failures:

    text_profile = {
        "profile_version": (
            "claim_extraction_text_profile_v1"
        ),

        "source_contract": (
            "claim_extraction_unit_contract_v1"
        ),

        "canonical_text_field": (
            canonical_text_field
        ),

        "eligible_records": len(
            eligible_document_ids
        ),

        "extraction_units": (
            materialized_unit_count
        ),

        "unit_counts_by_source_type": dict(
            sorted(
                unit_counts_by_source.items()
            )
        ),

        "source_text_profiles": (
            source_text_profiles
        ),

        "largest_units": (
            largest_units
        ),

        "measurement_contract": {
            "unit_measure": (
                "SUM_OF_RAW_TEXT_CHARACTERS_"
                "ACROSS_UNIT_DOCUMENTS"
            ),
            "prompt_markers_included": False,
            "tokenization_applied": False,
            "prompt_windowing_applied": False,
            "model_context_limit_assumed": False,
        },

        "current_state": {
            "canonical_text_field_resolved": True,
            "prompt_windowing_defined": False,
            "claim_extraction_started": False,
        },
    }


    text_profile_text = (
        json.dumps(
            text_profile,
            ensure_ascii=False,
            indent=2,
            sort_keys=True,
        )
        + "\n"
    )


    text_profile_sha256 = (
        hashlib.sha256(
            text_profile_text.encode(
                "utf-8"
            )
        ).hexdigest()
    )


    if text_profile_path.exists():

        existing_text = (
            text_profile_path.read_text(
                encoding="utf-8"
            )
        )

        assert (
            existing_text
            == text_profile_text
        ), (
            "Existing text profile differs "
            "from the validated staged profile. "
            "Refusing overwrite."
        )

        text_profile_write_status = (
            "REUSED_EXISTING_IDENTICAL_PROFILE"
        )

    else:

        with text_profile_path.open(
            "x",
            encoding="utf-8",
        ) as file:
            file.write(
                text_profile_text
            )

        text_profile_write_status = (
            "CREATED_NEW_PROFILE"
        )


    disk_text_profile = load_json(
        text_profile_path
    )

    assert (
        disk_text_profile
        == text_profile
    ), (
        "Disk text profile differs "
        "from staged profile."
    )


    disk_text_profile_sha256 = (
        file_sha256(
            text_profile_path
        )
    )

    assert (
        disk_text_profile_sha256
        == text_profile_sha256
    ), (
        "Text profile SHA-256 failed "
        "disk re-validation."
    )


text_profile_status = (
    "PASS"
    if not text_contract_failures
    else "FAIL"
)


print(
    "=== NB12i EXTRACTION TEXT PROFILE ==="
)

print()
print(
    "Eligible records:",
    len(
        eligible_document_ids
    ),
)

print(
    "Missing top-level text field:",
    len(
        missing_text_document_ids
    ),
)

print(
    "Empty/non-string text field:",
    len(
        empty_text_document_ids
    ),
)

print(
    "Canonical text field:",
    (
        canonical_text_field
        if canonical_text_field
        is not None
        else "UNRESOLVED"
    ),
)


if canonical_text_field is None:

    print()
    print(
        "Top-level string field candidates:"
    )

    for item in (
        candidate_string_fields[:12]
    ):
        print(
            "  -",
            item[
                "field"
            ],
            "| coverage:",
            item[
                "coverage"
            ],
            "| total chars:",
            item[
                "total_chars"
            ],
        )


else:

    print()
    print(
        "Materialized Extraction Units:",
        materialized_unit_count,
    )

    print()
    print(
        "Raw text size profiles:"
    )

    for source_type in sorted(
        source_text_profiles
    ):

        profile = (
            source_text_profiles[
                source_type
            ]
        )

        print()
        print(
            "  Source type:",
            source_type,
        )

        print(
            "    Units:",
            profile[
                "units"
            ],
        )

        print(
            "    Total chars:",
            profile[
                "total_raw_text_chars"
            ],
        )

        print(
            "    Min:",
            profile[
                "min_raw_text_chars"
            ],
        )

        print(
            "    P50:",
            profile[
                "p50_raw_text_chars"
            ],
        )

        print(
            "    P75:",
            profile[
                "p75_raw_text_chars"
            ],
        )

        print(
            "    P90:",
            profile[
                "p90_raw_text_chars"
            ],
        )

        print(
            "    P95:",
            profile[
                "p95_raw_text_chars"
            ],
        )

        print(
            "    P99:",
            profile[
                "p99_raw_text_chars"
            ],
        )

        print(
            "    Max:",
            profile[
                "max_raw_text_chars"
            ],
        )


    print()
    print(
        "Largest Extraction Units:"
    )

    for item in largest_units:
        print(
            "  -",
            item[
                "source_type"
            ],
            "| docs:",
            item[
                "document_count"
            ],
            "| chars:",
            item[
                "raw_text_chars"
            ],
            "|",
            item[
                "extraction_unit_id"
            ],
        )


print()
print(
    "Text contract failures:",
    len(
        text_contract_failures
    ),
)

for failure in (
    text_contract_failures
):
    print(
        "  -",
        failure,
    )


if text_profile_status == "PASS":

    print()
    print(
        "Text profile:",
        text_profile_path.relative_to(
            project_root
        ),
    )

    print(
        "Text profile SHA-256:",
        disk_text_profile_sha256,
    )

    print(
        "Write status:",
        text_profile_write_status,
    )


print()
print(
    "Prompt windowing defined:",
    False,
)

print(
    "Claim extraction started:",
    False,
)

print()
print(
    "CODE CELL 06 GATE:",
    text_profile_status,
)
```

<h2>Discovery قرارداد واقعی LLM برای Claim Extraction</h2>

<p>
Profile متن مرحلهٔ قبل نشان داد که Extraction Unitهای ساختاری از چند Character تا بیش از بیست‌وسه هزار Character اندازه دارند.
بنابراین قبل از ساخت Prompt Windowها باید Runtime واقعی LLM پروژه مشخص شود.
</p>

<p>
هدف این مرحله انتخاب مدل جدید یا انجام API Call نیست.
این مرحله فقط Notebookها و کدهای موجود پروژه را به‌صورت Read-Only بررسی می‌کند تا مشخص شود Pipelineهای قبلی Golden Dataset و Core-46 از چه Provider، Model، Endpoint و تنظیمات تولید استفاده کرده‌اند.
</p>

<h3>چرا از تنظیمات قبلی پروژه شروع می‌کنیم؟</h3>

<p>
Core-46 قبلاً مسیر Evidence-first با کیفیت مناسب‌تری ایجاد کرده است.
بنابراین Claim Registry نباید بدون بررسی Runtime واقعی پروژه یک Model Contract یا Prompt Limit جدید از روی حدس بسازد.
</p>

<h3>موارد مورد Discovery</h3>

<ul>
<li>نام متغیر یا مقدار Model مورد استفاده؛</li>
<li>Endpoint یا Client مورد استفاده؛</li>
<li>وجود تنظیمات <code>max_tokens</code> یا <code>max_completion_tokens</code>؛</li>
<li>Temperature و سایر Generation Parameterهای مرتبط؛</li>
<li>وجود هرگونه Context Limit یا Character Window صریح؛</li>
<li>الگوی واقعی API Call در Notebookهای مرتبط؛</li>
<li>وجود Retry یا Timeout Contract در Pipelineهای قبلی.</li>
</ul>

<h3>حفاظت از Secret</h3>

<p>
این مرحله هیچ API Key، Bearer Token یا Credential را چاپ نمی‌کند.
هر خطی که احتمال حمل Secret داشته باشد در Output Redact می‌شود.
</p>

<h3>اصل مهم</h3>

<p>
وجود <code>max_tokens</code> به معنی Context Window مدل نیست.
اگر Context Limit صریح در Repository پیدا نشود، این مرحله مقدار دلخواهی برای آن تولید نمی‌کند و وضعیت آن را <code>UNRESOLVED</code> ثبت می‌کند.
</p>

<h3>خروجی این مرحله</h3>

<p>
هیچ Artifact جدیدی ساخته نمی‌شود.
خروجی فقط Evidence خوانا از Runtime قبلی پروژه است تا مرحلهٔ بعد بتواند Model Context Contract و Prompt Windowing را Evidence-based تعریف کند.
</p>

<h3>معیار PASS</h3>

<ul>
<li>Notebookهای مرتبط قبلی به‌صورت Read-Only بررسی شوند؛</li>
<li>هیچ Secret در Output نمایش داده نشود؛</li>
<li>Evidence مربوط به Model / Endpoint / Generation Parameters در صورت وجود گزارش شود؛</li>
<li>وجود یا عدم وجود Context Limit صریح مشخص شود؛</li>
<li>هیچ API Call انجام نشود؛</li>
<li>هیچ فایل یا Artifact موجود تغییر نکند.</li>
</ul>

```python
# Code Cell 07

assert (
    text_profile_status
    == "PASS"
), (
    "Code Cell 06 must PASS before "
    "discovering the LLM runtime contract."
)


runtime_search_notebooks = sorted(
    path
    for path in related_notebooks
    if any(
        token in path.name.casefold()
        for token in [
            "12g",
            "12h",
            "13_",
            "golden",
            "quality",
            "rebuild",
        ]
    )
)


search_patterns = [
    "ARVAN_",
    "chat/completions",
    "max_tokens",
    "max_completion_tokens",
    "temperature",
    "timeout",
    "requests.post",
    "base_url",
    "OpenAI(",
    "context_window",
    "context_limit",
    "max_context",
    "CONTEXT_MAX",
    "MAX_CONTEXT",
    "MODEL_NAME",
    "CHAT_MODEL",
]


secret_markers = [
    "api_key",
    "authorization",
    "bearer ",
    "secret",
    "access_token",
]


def sanitize_runtime_line(line):
    lowered = line.casefold()

    if any(
        marker in lowered
        for marker in secret_markers
    ):
        return "<REDACTED SECRET-BEARING LINE>"

    return line.strip()


runtime_evidence = []

scanned_code_cells = 0


for notebook_path in (
    runtime_search_notebooks
):

    with notebook_path.open(
        "r",
        encoding="utf-8",
        errors="strict",
    ) as file:
        notebook_json = json.load(
            file
        )

    for cell_index, cell in enumerate(
        notebook_json.get(
            "cells",
            []
        ),
        1,
    ):

        if (
            cell.get(
                "cell_type"
            )
            != "code"
        ):
            continue

        scanned_code_cells += 1

        source = "".join(
            cell.get(
                "source",
                []
            )
        )

        for line_number, line in enumerate(
            source.splitlines(),
            1,
        ):

            lowered = line.casefold()

            matched_patterns = [
                pattern
                for pattern in search_patterns
                if pattern.casefold()
                in lowered
            ]

            if not matched_patterns:
                continue

            runtime_evidence.append({
                "notebook": str(
                    notebook_path.relative_to(
                        project_root
                    )
                ),
                "cell_index": (
                    cell_index
                ),
                "line_number": (
                    line_number
                ),
                "matched_patterns": (
                    matched_patterns
                ),
                "line": (
                    sanitize_runtime_line(
                        line
                    )
                ),
            })


model_evidence = [
    item
    for item in runtime_evidence
    if any(
        token.casefold()
        in item[
            "line"
        ].casefold()
        for token in [
            "CHAT_MODEL",
            "MODEL_NAME",
            "model",
        ]
    )
    and item[
        "line"
    ]
    != "<REDACTED SECRET-BEARING LINE>"
]


endpoint_evidence = [
    item
    for item in runtime_evidence
    if any(
        token.casefold()
        in item[
            "line"
        ].casefold()
        for token in [
            "chat/completions",
            "base_url",
            "CHAT_URL",
            "requests.post",
        ]
    )
]


generation_limit_evidence = [
    item
    for item in runtime_evidence
    if any(
        token.casefold()
        in item[
            "line"
        ].casefold()
        for token in [
            "max_tokens",
            "max_completion_tokens",
        ]
    )
]


temperature_evidence = [
    item
    for item in runtime_evidence
    if "temperature"
    in item[
        "line"
    ].casefold()
]


timeout_evidence = [
    item
    for item in runtime_evidence
    if "timeout"
    in item[
        "line"
    ].casefold()
]


explicit_context_limit_evidence = [
    item
    for item in runtime_evidence
    if any(
        token.casefold()
        in item[
            "line"
        ].casefold()
        for token in [
            "context_window",
            "context_limit",
            "max_context",
            "context_max",
            "max_context",
        ]
    )
]


runtime_discovery_failures = []


if not runtime_search_notebooks:
    runtime_discovery_failures.append(
        "No related notebooks were available "
        "for runtime discovery."
    )


if scanned_code_cells == 0:
    runtime_discovery_failures.append(
        "No code cells were available "
        "for runtime discovery."
    )


runtime_discovery_status = (
    "PASS"
    if not runtime_discovery_failures
    else "FAIL"
)


llm_runtime_discovery = {
    "notebooks_scanned": [
        str(
            path.relative_to(
                project_root
            )
        )
        for path in (
            runtime_search_notebooks
        )
    ],

    "code_cells_scanned": (
        scanned_code_cells
    ),

    "runtime_evidence_count": (
        len(
            runtime_evidence
        )
    ),

    "model_evidence_count": (
        len(
            model_evidence
        )
    ),

    "endpoint_evidence_count": (
        len(
            endpoint_evidence
        )
    ),

    "generation_limit_evidence_count": (
        len(
            generation_limit_evidence
        )
    ),

    "temperature_evidence_count": (
        len(
            temperature_evidence
        )
    ),

    "timeout_evidence_count": (
        len(
            timeout_evidence
        )
    ),

    "explicit_context_limit_evidence_count": (
        len(
            explicit_context_limit_evidence
        )
    ),

    "explicit_context_limit_status": (
        "FOUND"
        if explicit_context_limit_evidence
        else "UNRESOLVED"
    ),

    "api_call_performed": False,
    "files_modified": False,
}


print(
    "=== NB12i LLM RUNTIME DISCOVERY ==="
)

print()
print(
    "Notebooks scanned:",
    len(
        runtime_search_notebooks
    ),
)

for path in (
    runtime_search_notebooks
):
    print(
        "  -",
        path.relative_to(
            project_root
        ),
    )


print()
print(
    "Code cells scanned:",
    scanned_code_cells,
)

print(
    "Runtime evidence lines:",
    len(
        runtime_evidence
    ),
)

print()
print(
    "Model evidence:",
    len(
        model_evidence
    ),
)

print(
    "Endpoint / client evidence:",
    len(
        endpoint_evidence
    ),
)

print(
    "Generation-limit evidence:",
    len(
        generation_limit_evidence
    ),
)

print(
    "Temperature evidence:",
    len(
        temperature_evidence
    ),
)

print(
    "Timeout evidence:",
    len(
        timeout_evidence
    ),
)

print(
    "Explicit context-limit evidence:",
    len(
        explicit_context_limit_evidence
    ),
)

print(
    "Explicit context-limit status:",
    llm_runtime_discovery[
        "explicit_context_limit_status"
    ],
)


print()
print(
    "Relevant runtime evidence:"
)

for item in (
    runtime_evidence[:120]
):
    print(
        "  -",
        item[
            "notebook"
        ],
        f"| cell {item['cell_index']}",
        f"| line {item['line_number']}",
        "|",
        item[
            "line"
        ],
    )


if (
    len(
        runtime_evidence
    )
    > 120
):
    print(
        "  ...",
        len(
            runtime_evidence
        )
        - 120,
        "additional matching lines not printed",
    )


print()
print(
    "Secrets printed:",
    False,
)

print(
    "API call performed:",
    False,
)

print(
    "Files modified:",
    False,
)


print()
print(
    "Discovery failures:",
    len(
        runtime_discovery_failures
    ),
)

for failure in (
    runtime_discovery_failures
):
    print(
        "  -",
        failure,
    )


print()
print(
    "CODE CELL 07 GATE:",
    runtime_discovery_status,
)
```

<h2>تفکیک Model Context از Pipeline Window و تثبیت Runtime مرجع</h2>

<p>
Discovery مرحلهٔ قبل نشان داد که Notebookهای قبلی از Runtime مبتنی بر Arvan و Client سازگار با OpenAI استفاده کرده‌اند و Evidenceهای متعددی برای Model، Endpoint، Generation Limit، Temperature و Timeout وجود دارد.
</p>

<p>
با این حال، وجود عباراتی مانند <code>context_window</code> در کد لزوماً به معنی Context Window واقعی مدل نیست.
این عبارت ممکن است به Window داخلی Dataset، Prompt Builder یا Evaluation Pipeline اشاره کند.
</p>

<h3>هدف این مرحله</h3>

<p>
این مرحله فقط Evidenceهای مرتبط با Runtimeهای اخیر و موفق‌تر پروژه را دقیق‌تر بررسی می‌کند تا سه مفهوم از هم جدا شوند:
</p>

<ul>
<li><strong>Model Runtime:</strong> مدل و Provider واقعی مورد استفاده؛</li>
<li><strong>Generation Budget:</strong> مقادیری مانند <code>max_tokens</code> که برای خروجی مدل تعیین شده‌اند؛</li>
<li><strong>Input / Prompt Window Policy:</strong> محدودیت Character یا Context که خود Pipeline روی Evidence ورودی اعمال کرده است.</li>
</ul>

<h3>Notebookهای مرجع</h3>

<p>
تمرکز این مرحله بر Notebookهایی است که مستقیماً به Quality Rebuild، Core-46 و ادامهٔ Golden Dataset مربوط بوده‌اند:
</p>

<ul>
<li><code>12g_golden_dataset_quality_rebuild_lab.ipynb</code></li>
<li><code>12g_golden_dataset_rebuild_clean.ipynb</code></li>
<li><code>12h_pilot_and_expansion_audit.ipynb</code></li>
</ul>

<h3>چه چیزی استخراج می‌شود؟</h3>

<ul>
<li>مقدار واقعی Model در Outputهای ذخیره‌شده، اگر وجود داشته باشد؛</li>
<li>خطوط واقعی مربوط به <code>max_tokens</code>، <code>temperature</code> و <code>timeout</code>؛</li>
<li>تمام Candidateهای مربوط به Context / Window همراه با چند خط اطرافشان؛</li>
<li>Evidenceهای مربوط به Character Limit یا Prompt-size Policy قبلی؛</li>
<li>تفکیک Candidateهای Model Context از Pipeline Window.</li>
</ul>

<h3>اصل مهم</h3>

<p>
اگر Context Window واقعی مدل از Repository قابل اثبات نباشد، وضعیت آن <code>UNRESOLVED</code> باقی می‌ماند.
در آن صورت Windowing Claim Extraction بر اساس Pipeline Contract و اندازهٔ واقعی داده طراحی خواهد شد، نه بر اساس یک Context Limit ساختگی.
</p>

<h3>معیار PASS</h3>

<ul>
<li>Model Runtime قبلی در حد Evidence موجود مشخص شود؛</li>
<li>Generation Budgetهای قبلی قابل مشاهده باشند؛</li>
<li>Candidateهای Context به‌صورت معنایی بررسی شوند؛</li>
<li>Pipeline Window با Model Context اشتباه گرفته نشود؛</li>
<li>هیچ Secret چاپ نشود؛</li>
<li>هیچ API Call انجام نشود؛</li>
<li>هیچ Artifact یا فایل موجود تغییر نکند.</li>
</ul>

```python
# Code Cell 08

assert (
    runtime_discovery_status
    == "PASS"
), (
    "Code Cell 07 must PASS before "
    "resolving the runtime reference contract."
)


runtime_reference_names = {
    "12g_golden_dataset_quality_rebuild_lab.ipynb",
    "12g_golden_dataset_rebuild_clean.ipynb",
    "12h_pilot_and_expansion_audit.ipynb",
}


runtime_reference_notebooks = sorted(
    path
    for path in related_notebooks
    if path.name in runtime_reference_names
)


assert (
    len(runtime_reference_notebooks)
    == len(runtime_reference_names)
), (
    "Not all expected runtime reference "
    "notebooks were found."
)


runtime_keywords = [
    "ARVAN_CHAT_MODEL",
    "chat_model",
    "arvan_chat_model",
    "model=",
    '"model"',
    "max_tokens",
    "max_completion_tokens",
    "temperature",
    "timeout",
    "context_window",
    "context_limit",
    "context_max",
    "max_context",
    "max_chars",
    "context_chars",
    "char_limit",
    "window_chars",
    "prompt_window",
]


context_keywords = [
    "context_window",
    "context_limit",
    "context_max",
    "max_context",
    "max_chars",
    "context_chars",
    "char_limit",
    "window_chars",
    "prompt_window",
]


generation_keywords = [
    "max_tokens",
    "max_completion_tokens",
    "temperature",
    "timeout",
]


def sanitize_line(line):
    lowered = line.casefold()

    secret_markers = [
        "api_key",
        "authorization",
        "bearer ",
        "secret",
        "access_token",
    ]

    if any(
        marker in lowered
        for marker in secret_markers
    ):
        return "<REDACTED SECRET-BEARING LINE>"

    return line.rstrip()


def output_texts(cell):
    texts = []

    for output in cell.get(
        "outputs",
        []
    ):

        output_type = output.get(
            "output_type"
        )

        if output_type == "stream":

            text = output.get(
                "text",
                ""
            )

            if isinstance(
                text,
                list,
            ):
                text = "".join(
                    text
                )

            if text:
                texts.append(
                    str(text)
                )

        elif output_type in {
            "execute_result",
            "display_data",
        }:

            data = output.get(
                "data",
                {}
            )

            text = data.get(
                "text/plain"
            )

            if isinstance(
                text,
                list,
            ):
                text = "".join(
                    text
                )

            if text:
                texts.append(
                    str(text)
                )

    return texts


reference_code_evidence = []
reference_context_evidence = []
reference_generation_evidence = []
stored_output_evidence = []


for notebook_path in (
    runtime_reference_notebooks
):

    with notebook_path.open(
        "r",
        encoding="utf-8",
        errors="strict",
    ) as file:
        notebook_json = json.load(
            file
        )

    cells = notebook_json.get(
        "cells",
        []
    )

    for cell_index, cell in enumerate(
        cells,
        1,
    ):

        if cell.get(
            "cell_type"
        ) != "code":
            continue

        source_lines = "".join(
            cell.get(
                "source",
                []
            )
        ).splitlines()

        for line_index, line in enumerate(
            source_lines
        ):

            lowered = line.casefold()

            matched_runtime = [
                keyword
                for keyword in runtime_keywords
                if keyword.casefold()
                in lowered
            ]

            if matched_runtime:

                item = {
                    "notebook": str(
                        notebook_path.relative_to(
                            project_root
                        )
                    ),
                    "cell_index": (
                        cell_index
                    ),
                    "line_number": (
                        line_index + 1
                    ),
                    "line": (
                        sanitize_line(
                            line
                        )
                    ),
                }

                reference_code_evidence.append(
                    item
                )

            if any(
                keyword.casefold()
                in lowered
                for keyword in context_keywords
            ):

                start = max(
                    0,
                    line_index - 2,
                )

                end = min(
                    len(source_lines),
                    line_index + 3,
                )

                context_block = [
                    sanitize_line(
                        source_lines[
                            surrounding_index
                        ]
                    )
                    for surrounding_index in range(
                        start,
                        end,
                    )
                ]

                reference_context_evidence.append({
                    "notebook": str(
                        notebook_path.relative_to(
                            project_root
                        )
                    ),
                    "cell_index": (
                        cell_index
                    ),
                    "line_number": (
                        line_index + 1
                    ),
                    "context_block": (
                        context_block
                    ),
                })

            if any(
                keyword.casefold()
                in lowered
                for keyword in generation_keywords
            ):

                reference_generation_evidence.append({
                    "notebook": str(
                        notebook_path.relative_to(
                            project_root
                        )
                    ),
                    "cell_index": (
                        cell_index
                    ),
                    "line_number": (
                        line_index + 1
                    ),
                    "line": (
                        sanitize_line(
                            line
                        )
                    ),
                })


        for output_text in output_texts(
            cell
        ):

            for output_line in (
                output_text.splitlines()
            ):

                lowered = (
                    output_line.casefold()
                )

                if any(
                    marker in lowered
                    for marker in [
                        "chat model:",
                        "model:",
                        "context window",
                        "context limit",
                        "max tokens",
                        "max_tokens",
                        "temperature",
                        "timeout",
                    ]
                ):

                    stored_output_evidence.append({
                        "notebook": str(
                            notebook_path.relative_to(
                                project_root
                            )
                        ),
                        "cell_index": (
                            cell_index
                        ),
                        "line": (
                            sanitize_line(
                                output_line
                            )
                        ),
                    })


resolved_model_values = set()


for item in (
    stored_output_evidence
):

    line = item[
        "line"
    ].strip()

    lowered = line.casefold()

    if lowered.startswith(
        "chat model:"
    ):

        value = line.split(
            ":",
            1,
        )[1].strip()

        if value:
            resolved_model_values.add(
                value
            )


model_context_candidates = []

pipeline_context_candidates = []


for item in (
    reference_context_evidence
):

    joined = " ".join(
        item[
            "context_block"
        ]
    ).casefold()

    looks_like_model_context = (
        any(
            marker in joined
            for marker in [
                "model context",
                "context length",
                "context limit",
                "max context",
                "max_input_tokens",
                "context_window_tokens",
            ]
        )
        and any(
            marker in joined
            for marker in [
                "model",
                "token",
            ]
        )
    )

    if looks_like_model_context:
        model_context_candidates.append(
            item
        )

    else:
        pipeline_context_candidates.append(
            item
        )


runtime_reference_resolution = {
    "reference_notebooks": [
        str(
            path.relative_to(
                project_root
            )
        )
        for path in (
            runtime_reference_notebooks
        )
    ],

    "resolved_model_values": sorted(
        resolved_model_values
    ),

    "generation_evidence_count": (
        len(
            reference_generation_evidence
        )
    ),

    "context_candidate_count": (
        len(
            reference_context_evidence
        )
    ),

    "model_context_candidate_count": (
        len(
            model_context_candidates
        )
    ),

    "pipeline_context_candidate_count": (
        len(
            pipeline_context_candidates
        )
    ),

    "model_context_status": (
        "LOCAL_EVIDENCE_FOUND"
        if model_context_candidates
        else "UNRESOLVED"
    ),

    "api_call_performed": False,
    "files_modified": False,
}


runtime_reference_failures = []


if not reference_code_evidence:
    runtime_reference_failures.append(
        "No runtime evidence was found "
        "in the selected reference notebooks."
    )


if not reference_generation_evidence:
    runtime_reference_failures.append(
        "No generation-parameter evidence "
        "was found in the selected notebooks."
    )


runtime_reference_status = (
    "PASS"
    if not runtime_reference_failures
    else "FAIL"
)


print(
    "=== NB12i RUNTIME REFERENCE RESOLUTION ==="
)

print()
print(
    "Reference notebooks:",
    len(
        runtime_reference_notebooks
    ),
)

for path in (
    runtime_reference_notebooks
):
    print(
        "  -",
        path.relative_to(
            project_root
        ),
    )


print()
print(
    "Resolved model values from "
    "stored notebook outputs:",
    len(
        resolved_model_values
    ),
)

for value in sorted(
    resolved_model_values
):
    print(
        "  -",
        value,
    )


print()
print(
    "Generation-parameter evidence:",
    len(
        reference_generation_evidence
    ),
)

for item in (
    reference_generation_evidence[:40]
):
    print(
        "  -",
        item[
            "notebook"
        ],
        f"| cell {item['cell_index']}",
        f"| line {item['line_number']}",
        "|",
        item[
            "line"
        ],
    )


if (
    len(
        reference_generation_evidence
    )
    > 40
):
    print(
        "  ...",
        len(
            reference_generation_evidence
        )
        - 40,
        "additional generation lines not printed",
    )


print()
print(
    "Context/window candidates:",
    len(
        reference_context_evidence
    ),
)

print(
    "Model-context candidates:",
    len(
        model_context_candidates
    ),
)

print(
    "Pipeline-context candidates:",
    len(
        pipeline_context_candidates
    ),
)

print(
    "Model-context status:",
    runtime_reference_resolution[
        "model_context_status"
    ],
)


print()
print(
    "Context/window evidence:"
)

for item in (
    reference_context_evidence[:30]
):

    print(
        "  -",
        item[
            "notebook"
        ],
        f"| cell {item['cell_index']}",
        f"| around line {item['line_number']}",
    )

    for line in (
        item[
            "context_block"
        ]
    ):
        print(
            "      ",
            line,
        )


if (
    len(
        reference_context_evidence
    )
    > 30
):
    print(
        "  ...",
        len(
            reference_context_evidence
        )
        - 30,
        "additional context candidates not printed",
    )


print()
print(
    "Stored runtime outputs:"
)

for item in (
    stored_output_evidence[:30]
):
    print(
        "  -",
        item[
            "notebook"
        ],
        f"| cell {item['cell_index']}",
        "|",
        item[
            "line"
        ],
    )


if (
    len(
        stored_output_evidence
    )
    > 30
):
    print(
        "  ...",
        len(
            stored_output_evidence
        )
        - 30,
        "additional output lines not printed",
    )


print()
print(
    "Secrets printed:",
    False,
)

print(
    "API call performed:",
    False,
)

print(
    "Files modified:",
    False,
)


print()
print(
    "Resolution failures:",
    len(
        runtime_reference_failures
    ),
)

for failure in (
    runtime_reference_failures
):
    print(
        "  -",
        failure,
    )


print()
print(
    "CODE CELL 08 GATE:",
    runtime_reference_status,
)
```

<div style="direction: rtl; white-space: normal; line-height: 1.7;">

<h2>ساخت Prompt Windowهای Claim Extraction بر اساس Distribution واقعی Corpus</h2>

<p>
Runtime مرجع پروژه در مرحلهٔ قبل مشخص شد، اما Context Limit واقعی مدل از Repository قابل اثبات نبود.
بنابراین این Notebook هیچ Context Window ساختگی برای <code>Qwen3-30B-A3B</code> تعریف نمی‌کند.
</p>

<p>
به‌جای آن، Prompt Windowing به‌عنوان یک Contract داخلی Pipeline و مستقل از Maximum Context مدل طراحی می‌شود.
مقدار Budget مستقیماً از Distribution واقعی Extraction Unitهای مرحلهٔ قبل مشتق می‌شود.
</p>

<h3>مبنای انتخاب Window Budget</h3>

<p>
برای هر Source Type، صدک ۹۵ حجم Raw Text قبلاً اندازه‌گیری شده است.
Budget این مرحله برابر با بیشترین P95 میان Source Typeها خواهد بود:
</p>

<pre><code>Window Budget
=
max(
    Book P95,
    Transcript P95,
    Website P95
)
</code></pre>

<p>
بنابراین هیچ عدد ثابت دلخواه در Pipeline وارد نمی‌شود.
این Budget فقط یک محدودیت داخلی برای Materialization Prompt است و به معنی Context Limit واقعی مدل نیست.
</p>

<h3>حفظ ساختار و Evidence</h3>

<p>
Prompt Windowها متن Source را در Artifact جدید کپی نمی‌کنند.
هر Window فقط Sliceهایی از Document اصلی را با این اطلاعات نگه می‌دارد:
</p>

<ul>
<li><code>document_id</code></li>
<li><code>char_start</code></li>
<li><code>char_end</code></li>
<li>نوع Segment؛</li>
<li>اینکه Slice اصلی است یا Overlap.</li>
</ul>

<p>
در زمان Claim Extraction، متن دقیق از Accepted Corpus براساس همین Spanها Materialize خواهد شد.
</p>

<h3>قواعد Windowing</h3>

<ul>
<li>هیچ Window از Extraction Unit خود خارج نمی‌شود؛</li>
<li>Source، Authority و Claim Discovery Role ثابت می‌مانند؛</li>
<li>Recordهای کوچک تا حد Budget در کنار هم قرار می‌گیرند؛</li>
<li>Record بزرگ‌تر از Budget ابتدا براساس Boundaryهای متنی شکسته می‌شود؛</li>
<li>اگر یک Segment هنوز بزرگ‌تر از Budget باشد، Hard Character Split با Offset واقعی انجام می‌شود؛</li>
<li>در صورت امکان، آخرین Segment Window قبلی به‌عنوان یک Overlap Slice در Window بعدی تکرار می‌شود؛</li>
<li>Overlap فقط برای حفظ Context مرزی است و در Coverage اصلی دوباره شمرده نمی‌شود.</li>
</ul>

<h3>Coverage Contract</h3>

<p>
تمام متن هر ۱۵٬۱۵۸ Record مجاز باید در Sliceهای اصلی دقیقاً یک‌بار پوشش داده شود.
Overlap مجاز است، اما نباید باعث Gap یا Double-count در Primary Coverage شود.
</p>

<p>
برای هر Document بررسی خواهد شد که Primary Spanها دقیقاً از Character صفر تا انتهای <code>text</code> را بدون Gap و بدون Overlap پوشش دهند.
</p>

<h3>Artifactهای این مرحله</h3>

<ul>
<li><code>claim_extraction_prompt_window_contract_v1.json</code></li>
<li><code>claim_extraction_prompt_windows_v1.jsonl</code></li>
</ul>

<p>
این Artifactها هنوز هیچ Claim تولید نمی‌کنند.
آن‌ها فقط Population نهایی Promptهای Claim Extraction را تعریف می‌کنند.
</p>

<h3>معیار PASS</h3>

<ul>
<li>Window Budget فقط از Profile واقعی مرحلهٔ قبل مشتق شده باشد؛</li>
<li>تمام ۲٬۱۲۹ Structural Unit حداقل یک Prompt Window داشته باشند؛</li>
<li>تمام ۱۵٬۱۵۸ Document در Primary Coverage حضور داشته باشند؛</li>
<li>Primary Character Coverage دقیقاً با کل متن Corpus مجاز برابر باشد؛</li>
<li>هیچ Gap یا Primary Overlap در Document Spanها وجود نداشته باشد؛</li>
<li>هیچ Window از Budget تعریف‌شده بزرگ‌تر نباشد؛</li>
<li>هیچ Window Source، Authority یا Role مختلط نداشته باشد؛</li>
<li>Overlap از Primary Coverage جدا باشد؛</li>
<li>Artifactهای موجود Overwrite نشوند؛</li>
<li>هر دو Artifact پس از Publication از Disk دوباره Validate شوند؛</li>
<li>هیچ API یا LLM Call انجام نشود.</li>
</ul>

</div>

```python
# Code Cell 09

assert (
    runtime_reference_status
    == "PASS"
), (
    "Code Cell 08 must PASS before "
    "building prompt windows."
)

assert (
    text_profile_status
    == "PASS"
), (
    "Code Cell 06 must PASS before "
    "building prompt windows."
)


prompt_window_contract_path = (
    registry_artifact_root
    / "claim_extraction_prompt_window_contract_v1.json"
)

prompt_window_manifest_path = (
    registry_artifact_root
    / "claim_extraction_prompt_windows_v1.jsonl"
)


source_p95_values = {
    source_type: profile[
        "p95_raw_text_chars"
    ]
    for source_type, profile in (
        text_profile[
            "source_text_profiles"
        ].items()
    )
}


window_char_budget = max(
    source_p95_values.values()
)


assert (
    window_char_budget > 0
), (
    "Derived Prompt Window budget "
    "must be positive."
)


def split_long_text_into_spans(
    text,
    max_chars,
):
    assert isinstance(
        text,
        str,
    )

    assert max_chars > 0

    if len(text) <= max_chars:
        return [
            {
                "char_start": 0,
                "char_end": len(text),
                "segment_kind": (
                    "FULL_DOCUMENT"
                ),
            }
        ]


    boundary_chars = {
        ".",
        "!",
        "?",
        "؟",
        "\n",
    }

    candidate_spans = []

    span_start = 0


    for index, char in enumerate(
        text
    ):

        if char not in boundary_chars:
            continue

        span_end = index + 1

        if span_end > span_start:
            candidate_spans.append(
                (
                    span_start,
                    span_end,
                )
            )

            span_start = (
                span_end
            )


    if span_start < len(text):
        candidate_spans.append(
            (
                span_start,
                len(text),
            )
        )


    if not candidate_spans:
        candidate_spans = [
            (
                0,
                len(text),
            )
        ]


    final_spans = []


    for (
        candidate_start,
        candidate_end,
    ) in candidate_spans:

        candidate_length = (
            candidate_end
            - candidate_start
        )

        if (
            candidate_length
            <= max_chars
        ):

            final_spans.append({
                "char_start": (
                    candidate_start
                ),
                "char_end": (
                    candidate_end
                ),
                "segment_kind": (
                    "TEXT_BOUNDARY_SEGMENT"
                ),
            })

            continue


        split_start = (
            candidate_start
        )


        while (
            split_start
            < candidate_end
        ):

            split_end = min(
                split_start
                + max_chars,
                candidate_end,
            )

            final_spans.append({
                "char_start": (
                    split_start
                ),
                "char_end": (
                    split_end
                ),
                "segment_kind": (
                    "HARD_CHAR_SPLIT"
                ),
            })

            split_start = (
                split_end
            )


    previous_end = 0


    for span in final_spans:

        assert (
            span[
                "char_start"
            ]
            == previous_end
        ), (
            "Internal text segmentation "
            "created a gap or overlap."
        )

        assert (
            span[
                "char_end"
            ]
            > span[
                "char_start"
            ]
        ), (
            "Internal text segmentation "
            "created an empty span."
        )

        assert (
            span[
                "char_end"
            ]
            - span[
                "char_start"
            ]
            <= max_chars
        ), (
            "Internal text segment exceeds "
            "the Prompt Window budget."
        )

        previous_end = (
            span[
                "char_end"
            ]
        )


    assert (
        previous_end
        == len(text)
    ), (
        "Internal text segmentation did "
        "not cover the full document."
    )


    return final_spans


def build_unit_segments(
    unit,
):
    segments = []


    for document_id in (
        unit[
            "document_ids"
        ]
    ):

        record = (
            accepted_records_by_id[
                document_id
            ]
        )

        text = record[
            canonical_text_field
        ]

        spans = (
            split_long_text_into_spans(
                text,
                window_char_budget,
            )
        )


        for span_index, span in enumerate(
            spans,
            1,
        ):

            segment_chars = (
                span[
                    "char_end"
                ]
                - span[
                    "char_start"
                ]
            )

            segments.append({
                "document_id": (
                    document_id
                ),
                "char_start": (
                    span[
                        "char_start"
                    ]
                ),
                "char_end": (
                    span[
                        "char_end"
                    ]
                ),
                "segment_chars": (
                    segment_chars
                ),
                "segment_kind": (
                    span[
                        "segment_kind"
                    ]
                ),
                "document_segment_index": (
                    span_index
                ),
            })


    return segments


def stable_window_id(
    extraction_unit_id,
    window_index,
    slices,
):
    material = stable_json({
        "extraction_unit_id": (
            extraction_unit_id
        ),
        "window_index": (
            window_index
        ),
        "slices": [
            {
                "document_id": (
                    item[
                        "document_id"
                    ]
                ),
                "char_start": (
                    item[
                        "char_start"
                    ]
                ),
                "char_end": (
                    item[
                        "char_end"
                    ]
                ),
                "is_overlap": (
                    item[
                        "is_overlap"
                    ]
                ),
            }
            for item in slices
        ],
    })

    digest = hashlib.sha256(
        material.encode(
            "utf-8"
        )
    ).hexdigest()[:24]

    return (
        f"claim-window:{digest}"
    )


def finalize_window(
    unit,
    window_index,
    slices,
):
    assert slices

    raw_text_chars = sum(
        item[
            "segment_chars"
        ]
        for item in slices
    )

    primary_text_chars = sum(
        item[
            "segment_chars"
        ]
        for item in slices
        if not item[
            "is_overlap"
        ]
    )


    assert (
        raw_text_chars
        <= window_char_budget
    ), (
        "Prompt Window exceeds "
        "derived Character budget."
    )


    return {
        "prompt_window_id": (
            stable_window_id(
                unit[
                    "extraction_unit_id"
                ],
                window_index,
                slices,
            )
        ),

        "window_version": (
            "claim_extraction_prompt_window_v1"
        ),

        "extraction_unit_id": (
            unit[
                "extraction_unit_id"
            ]
        ),

        "window_index": (
            window_index
        ),

        "source_type": (
            unit[
                "source_type"
            ]
        ),

        "source_id": (
            unit[
                "source_id"
            ]
        ),

        "authority_class": (
            unit[
                "authority_class"
            ]
        ),

        "claim_discovery_role": (
            unit[
                "claim_discovery_role"
            ]
        ),

        "slice_count": len(
            slices
        ),

        "primary_slice_count": sum(
            1
            for item in slices
            if not item[
                "is_overlap"
            ]
        ),

        "overlap_slice_count": sum(
            1
            for item in slices
            if item[
                "is_overlap"
            ]
        ),

        "raw_text_chars": (
            raw_text_chars
        ),

        "primary_text_chars": (
            primary_text_chars
        ),

        "slices": [
            {
                "document_id": (
                    item[
                        "document_id"
                    ]
                ),
                "char_start": (
                    item[
                        "char_start"
                    ]
                ),
                "char_end": (
                    item[
                        "char_end"
                    ]
                ),
                "segment_kind": (
                    item[
                        "segment_kind"
                    ]
                ),
                "document_segment_index": (
                    item[
                        "document_segment_index"
                    ]
                ),
                "is_overlap": (
                    item[
                        "is_overlap"
                    ]
                ),
            }
            for item in slices
        ],
    }


prompt_windows = []

split_unit_count = 0
windows_with_overlap = 0


for unit in units:

    segments = build_unit_segments(
        unit
    )

    assert segments, (
        "Extraction Unit produced "
        "no Prompt segments: "
        f"{unit['extraction_unit_id']}"
    )


    unit_windows = []

    current_slices = []
    current_chars = 0


    for segment in segments:

        segment_chars = (
            segment[
                "segment_chars"
            ]
        )

        assert (
            segment_chars
            <= window_char_budget
        )


        if (
            current_slices
            and (
                current_chars
                + segment_chars
                > window_char_budget
            )
        ):

            unit_windows.append(
                finalize_window(
                    unit,
                    len(
                        unit_windows
                    ) + 1,
                    current_slices,
                )
            )


            previous_primary_segment = None


            for previous_slice in reversed(
                current_slices
            ):

                if not previous_slice[
                    "is_overlap"
                ]:
                    previous_primary_segment = (
                        previous_slice
                    )
                    break


            current_slices = []
            current_chars = 0


            if (
                previous_primary_segment
                is not None
                and (
                    previous_primary_segment[
                        "segment_chars"
                    ]
                    + segment_chars
                    <= window_char_budget
                )
            ):

                overlap_slice = dict(
                    previous_primary_segment
                )

                overlap_slice[
                    "is_overlap"
                ] = True

                current_slices.append(
                    overlap_slice
                )

                current_chars += (
                    overlap_slice[
                        "segment_chars"
                    ]
                )


        primary_slice = dict(
            segment
        )

        primary_slice[
            "is_overlap"
        ] = False

        current_slices.append(
            primary_slice
        )

        current_chars += (
            segment_chars
        )


    if current_slices:

        unit_windows.append(
            finalize_window(
                unit,
                len(
                    unit_windows
                ) + 1,
                current_slices,
            )
        )


    if len(
        unit_windows
    ) > 1:
        split_unit_count += 1


    windows_with_overlap += sum(
        1
        for window in unit_windows
        if window[
            "overlap_slice_count"
        ] > 0
    )


    prompt_windows.extend(
        unit_windows
    )


prompt_windows = sorted(
    prompt_windows,
    key=lambda item: (
        item[
            "source_type"
        ],
        item[
            "source_id"
        ],
        item[
            "extraction_unit_id"
        ],
        item[
            "window_index"
        ],
    ),
)


prompt_window_ids = [
    item[
        "prompt_window_id"
    ]
    for item in prompt_windows
]


assert (
    len(
        prompt_window_ids
    )
    == len(
        set(
            prompt_window_ids
        )
    )
), (
    "Duplicate prompt_window_id detected."
)


windows_by_unit = Counter(
    item[
        "extraction_unit_id"
    ]
    for item in prompt_windows
)


units_without_windows = [
    unit[
        "extraction_unit_id"
    ]
    for unit in units
    if windows_by_unit[
        unit[
            "extraction_unit_id"
        ]
    ] == 0
]


assert not units_without_windows, (
    "At least one Extraction Unit "
    "has no Prompt Window."
)


window_over_budget_count = sum(
    1
    for item in prompt_windows
    if item[
        "raw_text_chars"
    ]
    > window_char_budget
)


assert (
    window_over_budget_count
    == 0
), (
    "At least one Prompt Window "
    "exceeds the Character budget."
)


primary_spans_by_document = defaultdict(
    list
)

overlap_span_count = 0


for window in prompt_windows:

    for item in window[
        "slices"
    ]:

        span = (
            item[
                "char_start"
            ],
            item[
                "char_end"
            ],
        )

        if item[
            "is_overlap"
        ]:

            overlap_span_count += 1

        else:

            primary_spans_by_document[
                item[
                    "document_id"
                ]
            ].append(
                span
            )


documents_missing_primary_coverage = []

documents_with_primary_gap = []

documents_with_primary_overlap = []

documents_with_invalid_end = []


for document_id in sorted(
    eligible_document_ids
):

    text = (
        accepted_records_by_id[
            document_id
        ][
            canonical_text_field
        ]
    )

    spans = sorted(
        primary_spans_by_document.get(
            document_id,
            [],
        )
    )


    if not spans:

        documents_missing_primary_coverage.append(
            document_id
        )

        continue


    previous_end = 0


    for (
        char_start,
        char_end,
    ) in spans:

        if char_start > previous_end:
            documents_with_primary_gap.append(
                document_id
            )

            break

        if char_start < previous_end:
            documents_with_primary_overlap.append(
                document_id
            )

            break

        previous_end = (
            char_end
        )


    if (
        previous_end
        != len(text)
    ):
        documents_with_invalid_end.append(
            document_id
        )


assert not (
    documents_missing_primary_coverage
), (
    "Some eligible Documents have "
    "no Primary Prompt coverage."
)

assert not (
    documents_with_primary_gap
), (
    "Primary Prompt coverage contains "
    "Document gaps."
)

assert not (
    documents_with_primary_overlap
), (
    "Primary Prompt coverage contains "
    "Document overlaps."
)

assert not (
    documents_with_invalid_end
), (
    "Primary Prompt coverage does not "
    "reach the full Document text."
)


expected_total_primary_chars = sum(
    len(
        accepted_records_by_id[
            document_id
        ][
            canonical_text_field
        ]
    )
    for document_id in (
        eligible_document_ids
    )
)


actual_total_primary_chars = sum(
    window[
        "primary_text_chars"
    ]
    for window in prompt_windows
)


assert (
    actual_total_primary_chars
    == expected_total_primary_chars
), (
    "Total Primary Prompt Character "
    "coverage does not match "
    "eligible Corpus text."
)


unit_ids_from_windows = {
    item[
        "extraction_unit_id"
    ]
    for item in prompt_windows
}


assert (
    unit_ids_from_windows
    == set(
        unit_ids
    )
), (
    "Prompt Window Manifest does not "
    "cover exactly the Extraction Unit set."
)


window_counts_by_source = Counter(
    item[
        "source_type"
    ]
    for item in prompt_windows
)


window_char_values_by_source = defaultdict(
    list
)


for item in prompt_windows:

    window_char_values_by_source[
        item[
            "source_type"
        ]
    ].append(
        item[
            "raw_text_chars"
        ]
    )


window_profiles_by_source = {}


for source_type in sorted(
    window_char_values_by_source
):

    values = sorted(
        window_char_values_by_source[
            source_type
        ]
    )

    window_profiles_by_source[
        source_type
    ] = {
        "windows": len(
            values
        ),
        "min_chars": (
            values[0]
        ),
        "p50_chars": (
            profile_quantile(
                values,
                0.50,
            )
        ),
        "p90_chars": (
            profile_quantile(
                values,
                0.90,
            )
        ),
        "p95_chars": (
            profile_quantile(
                values,
                0.95,
            )
        ),
        "p99_chars": (
            profile_quantile(
                values,
                0.99,
            )
        ),
        "max_chars": (
            values[-1]
        ),
    }


prompt_window_contract = {
    "contract_version": (
        "claim_extraction_prompt_window_contract_v1"
    ),

    "source_contract": (
        "claim_extraction_unit_contract_v1"
    ),

    "source_text_profile": (
        "claim_extraction_text_profile_v1"
    ),

    "runtime_reference_model": (
        sorted(
            resolved_model_values
        )[0]
        if len(
            resolved_model_values
        ) == 1
        else "UNRESOLVED"
    ),

    "model_context_limit_status": (
        "UNRESOLVED"
    ),

    "window_budget": {
        "unit": (
            "RAW_TEXT_CHARACTERS"
        ),

        "derivation": (
            "MAX_SOURCE_P95_RAW_TEXT_CHARS"
        ),

        "source_p95_values": (
            source_p95_values
        ),

        "max_raw_text_chars": (
            window_char_budget
        ),

        "represents_model_context_limit": (
            False
        ),
    },

    "segmentation_policy": {
        "primary_boundary": (
            "DOCUMENT_BOUNDARY"
        ),

        "oversized_document_policy": (
            "TEXT_BOUNDARY_THEN_HARD_CHAR_SPLIT"
        ),

        "overlap_policy": (
            "ONE_PREVIOUS_PRIMARY_SEGMENT_WHEN_"
            "IT_FITS_WITHIN_WINDOW_BUDGET"
        ),

        "overlap_counts_as_primary_coverage": (
            False
        ),
    },

    "coverage_policy": {
        "eligible_documents_must_be_covered": True,
        "primary_coverage_must_be_gap_free": True,
        "primary_coverage_must_not_overlap": True,
        "full_document_text_must_be_covered": True,
    },

    "retrieval_chunk_dependency": False,

    "claim_extraction_started": False,

    "manifest": (
        "claim_extraction_prompt_windows_v1.jsonl"
    ),
}


prompt_window_contract_text = (
    json.dumps(
        prompt_window_contract,
        ensure_ascii=False,
        indent=2,
        sort_keys=True,
    )
    + "\n"
)


prompt_window_manifest_text = "".join(
    json.dumps(
        item,
        ensure_ascii=False,
        sort_keys=True,
    )
    + "\n"
    for item in prompt_windows
)


prompt_window_contract_sha256 = (
    hashlib.sha256(
        prompt_window_contract_text.encode(
            "utf-8"
        )
    ).hexdigest()
)


prompt_window_manifest_sha256 = (
    hashlib.sha256(
        prompt_window_manifest_text.encode(
            "utf-8"
        )
    ).hexdigest()
)


if prompt_window_contract_path.exists():

    existing_text = (
        prompt_window_contract_path.read_text(
            encoding="utf-8"
        )
    )

    assert (
        existing_text
        == prompt_window_contract_text
    ), (
        "Existing Prompt Window Contract "
        "differs from the validated staged "
        "contract. Refusing overwrite."
    )

    prompt_window_contract_write_status = (
        "REUSED_EXISTING_IDENTICAL_CONTRACT"
    )

else:

    with prompt_window_contract_path.open(
        "x",
        encoding="utf-8",
    ) as file:
        file.write(
            prompt_window_contract_text
        )

    prompt_window_contract_write_status = (
        "CREATED_NEW_CONTRACT"
    )


if prompt_window_manifest_path.exists():

    existing_text = (
        prompt_window_manifest_path.read_text(
            encoding="utf-8"
        )
    )

    assert (
        existing_text
        == prompt_window_manifest_text
    ), (
        "Existing Prompt Window Manifest "
        "differs from the validated staged "
        "manifest. Refusing overwrite."
    )

    prompt_window_manifest_write_status = (
        "REUSED_EXISTING_IDENTICAL_MANIFEST"
    )

else:

    with prompt_window_manifest_path.open(
        "x",
        encoding="utf-8",
    ) as file:
        file.write(
            prompt_window_manifest_text
        )

    prompt_window_manifest_write_status = (
        "CREATED_NEW_MANIFEST"
    )


disk_prompt_window_contract = load_json(
    prompt_window_contract_path
)

disk_prompt_windows = load_jsonl(
    prompt_window_manifest_path
)


assert (
    disk_prompt_window_contract
    == prompt_window_contract
), (
    "Disk Prompt Window Contract differs "
    "from staged Contract."
)

assert (
    disk_prompt_windows
    == prompt_windows
), (
    "Disk Prompt Window Manifest differs "
    "from staged Manifest."
)


disk_prompt_window_contract_sha256 = (
    file_sha256(
        prompt_window_contract_path
    )
)

disk_prompt_window_manifest_sha256 = (
    file_sha256(
        prompt_window_manifest_path
    )
)


assert (
    disk_prompt_window_contract_sha256
    == prompt_window_contract_sha256
)

assert (
    disk_prompt_window_manifest_sha256
    == prompt_window_manifest_sha256
)


prompt_window_status = "PASS"


print(
    "=== NB12i CLAIM EXTRACTION PROMPT WINDOWS ==="
)

print()
print(
    "Runtime reference model:",
    prompt_window_contract[
        "runtime_reference_model"
    ],
)

print(
    "Model context limit:",
    "UNRESOLVED",
)

print()
print(
    "Source P95 raw-text chars:",
    source_p95_values,
)

print(
    "Derived Prompt Window budget:",
    window_char_budget,
)

print(
    "Budget represents model context limit:",
    False,
)


print()
print(
    "Structural Extraction Units:",
    len(
        units
    ),
)

print(
    "Prompt Windows:",
    len(
        prompt_windows
    ),
)

print(
    "Units split into multiple windows:",
    split_unit_count,
)

print(
    "Windows containing overlap:",
    windows_with_overlap,
)

print(
    "Overlap slices:",
    overlap_span_count,
)


print()
print(
    "Windows by source type:",
    dict(
        sorted(
            window_counts_by_source.items()
        )
    ),
)


print()
print(
    "Prompt Window size profiles:"
)

for source_type in sorted(
    window_profiles_by_source
):

    print(
        "  -",
        source_type,
        window_profiles_by_source[
            source_type
        ],
    )


print()
print(
    "Window over-budget count:",
    window_over_budget_count,
)

print(
    "Documents missing Primary coverage:",
    len(
        documents_missing_primary_coverage
    ),
)

print(
    "Documents with Primary gaps:",
    len(
        documents_with_primary_gap
    ),
)

print(
    "Documents with Primary overlaps:",
    len(
        documents_with_primary_overlap
    ),
)

print(
    "Documents with invalid final span:",
    len(
        documents_with_invalid_end
    ),
)


print()
print(
    "Expected Primary chars:",
    expected_total_primary_chars,
)

print(
    "Actual Primary chars:",
    actual_total_primary_chars,
)

print(
    "Primary Character coverage exact:",
    (
        expected_total_primary_chars
        == actual_total_primary_chars
    ),
)


print()
print(
    "Prompt Window Contract:",
    prompt_window_contract_path.relative_to(
        project_root
    ),
)

print(
    "Contract SHA-256:",
    disk_prompt_window_contract_sha256,
)

print(
    "Contract write status:",
    prompt_window_contract_write_status,
)


print()
print(
    "Prompt Window Manifest:",
    prompt_window_manifest_path.relative_to(
        project_root
    ),
)

print(
    "Manifest SHA-256:",
    disk_prompt_window_manifest_sha256,
)

print(
    "Manifest write status:",
    prompt_window_manifest_write_status,
)


print()
print(
    "API call performed:",
    False,
)

print(
    "Claim extraction started:",
    False,
)


print()
print(
    "CODE CELL 09 GATE:",
    prompt_window_status,
)
```

</p>

<h3>اصل Evidence-First</h3>

<p>
مدل اجازه ندارد ابتدا یک گزاره از دانش عمومی خود بسازد و بعد برای آن Evidence پیدا کند.
هر Claim باید مستقیماً از متن Window استخراج شود و حداقل یک Quote دقیق از Source داشته باشد.
</p>

<pre><code>Prompt Window
↓
Exact Source Evidence
↓
Atomic Claim
↓
Raw Claim
</code></pre>

<h3>در این مرحله Question ساخته نمی‌شود</h3>

<p>
هدف Claim Registry ساخت واحدهای دانشی است.
Question و Answer متعلق به Golden Dataset هستند و در این مرحله تولید نمی‌شوند.
</p>

<h3>طراحی Pilot</h3>

<p>
Pilot به‌صورت Deterministic از سه Source Type انتخاب می‌شود:
</p>

<ul>
<li>۵ Window از Book؛</li>
<li>۵ Window از Transcript؛</li>
<li>۵ Window از Website؛</li>
</ul>

<p>
Website به‌طور جداگانه Primary و Secondary را پوشش خواهد داد تا Attribution ثانویه نیز در همان Pilot آزمایش شود.
انتخاب Windowها بر اساس Quantileهای واقعی اندازهٔ Prompt انجام می‌شود تا فقط Windowهای خیلی کوچک یا فقط Windowهای خیلی بزرگ بررسی نشوند.
</p>

<h3>قواعد تولید Claim</h3>

<ul>
<li>حداکثر ۴ Claim از هر Prompt Window؛</li>
<li>Claim باید یک گزارهٔ مستقل و تا حد ممکن Atomic باشد؛</li>
<li>Claim باید به فارسی نوشته شود؛</li>
<li>هیچ Question یا Answer تولید نشود؛</li>
<li>هر Claim باید حداقل یک Evidence داشته باشد؛</li>
<li>Evidence فقط از Sliceهای Primary قابل انتخاب است؛</li>
<li><code>quote_exact</code> باید عیناً و بدون بازنویسی در Slice مربوط وجود داشته باشد؛</li>
<li>Quote باید داخل همان Slice یکتا باشد تا Locator دقیق آن قابل محاسبه باشد؛</li>
<li>اگر Window Claim مستقل و مفیدی ندارد، مدل مجاز است لیست خالی برگرداند؛</li>
<li>دانش بیرون از متن ممنوع است.</li>
</ul>

```python
# Code Cell 10

import os
import re
import time

from openai import OpenAI


assert (
    prompt_window_status
    == "PASS"
), (
    "Code Cell 09 must PASS before "
    "running the Claim Extraction Pilot."
)


PILOT_VERSION = (
    "claim_extraction_pilot_v1"
)

MAX_CLAIMS_PER_WINDOW = 4
MAX_EVIDENCE_ITEMS_PER_CLAIM = 3
PILOT_MAX_TOKENS = 2500
PILOT_TEMPERATURE = 0
PILOT_TIMEOUT_SECONDS = 180.0
PILOT_MAX_ATTEMPTS = 3


ALLOWED_CLAIM_TYPES = {
    "DEFINITION",
    "RELATION",
    "CAUSAL",
    "RULE_OR_PRINCIPLE",
    "PROCEDURE_OR_ACTION",
    "COMPARISON",
    "CONDITION_OR_CONSTRAINT",
    "CLASSIFICATION",
    "NUMERIC_OR_QUANTITATIVE",
    "OTHER_ATOMIC",
}


pilot_raw_path = (
    registry_artifact_root
    / "claim_extraction_pilot_raw_v1.jsonl"
)

pilot_evidence_path = (
    registry_artifact_root
    / "claim_extraction_pilot_evidence_v1.jsonl"
)

pilot_report_path = (
    registry_artifact_root
    / "claim_extraction_pilot_report_v1.json"
)

pilot_cache_root = (
    registry_artifact_root
    / "claim_extraction_pilot_cache_v1"
)


def discover_arvan_env_names():
    discovered = set()

    for notebook_path in (
        runtime_reference_notebooks
    ):

        with notebook_path.open(
            "r",
            encoding="utf-8",
            errors="strict",
        ) as file:
            notebook_json = json.load(
                file
            )

        for cell in notebook_json.get(
            "cells",
            []
        ):

            if (
                cell.get(
                    "cell_type"
                )
                != "code"
            ):
                continue

            source = "".join(
                cell.get(
                    "source",
                    []
                )
            )

            matches = re.findall(
                r"""os\.getenv\(\s*["']([^"']+)["']""",
                source,
            )

            for name in matches:

                if (
                    "ARVAN"
                    in name.upper()
                ):
                    discovered.add(
                        name
                    )

    return sorted(
        discovered
    )


arvan_env_names = (
    discover_arvan_env_names()
)


api_key_env_candidates = [
    name
    for name in arvan_env_names
    if "KEY" in name.upper()
]

chat_url_env_candidates = [
    name
    for name in arvan_env_names
    if (
        "CHAT" in name.upper()
        and "URL" in name.upper()
    )
]

chat_model_env_candidates = [
    name
    for name in arvan_env_names
    if (
        "CHAT" in name.upper()
        and "MODEL" in name.upper()
    )
]


assert (
    len(
        api_key_env_candidates
    )
    == 1
), (
    "Could not resolve exactly one "
    "Arvan API-key environment variable. "
    f"Candidates: {api_key_env_candidates}"
)

assert (
    len(
        chat_url_env_candidates
    )
    == 1
), (
    "Could not resolve exactly one "
    "Arvan Chat URL environment variable. "
    f"Candidates: {chat_url_env_candidates}"
)

assert (
    len(
        chat_model_env_candidates
    )
    == 1
), (
    "Could not resolve exactly one "
    "Arvan Chat Model environment variable. "
    f"Candidates: {chat_model_env_candidates}"
)


api_key_env_name = (
    api_key_env_candidates[0]
)

chat_url_env_name = (
    chat_url_env_candidates[0]
)

chat_model_env_name = (
    chat_model_env_candidates[0]
)


pilot_api_key = (
    os.getenv(
        api_key_env_name
    )
    or ""
).strip()

pilot_chat_url = (
    os.getenv(
        chat_url_env_name
    )
    or ""
).strip()

pilot_chat_model = (
    os.getenv(
        chat_model_env_name
    )
    or ""
).strip()


assert pilot_api_key, (
    f"{api_key_env_name} is missing."
)

assert pilot_chat_url, (
    f"{chat_url_env_name} is missing."
)

assert pilot_chat_model, (
    f"{chat_model_env_name} is missing."
)


assert (
    len(
        resolved_model_values
    )
    == 1
), (
    "Reference runtime model must "
    "be uniquely resolved."
)


reference_model = sorted(
    resolved_model_values
)[0]


assert (
    pilot_chat_model
    == reference_model
), (
    "Configured Chat model differs from "
    "the validated reference runtime. "
    f"Configured={pilot_chat_model!r}, "
    f"Reference={reference_model!r}"
)


pilot_client = OpenAI(
    api_key=pilot_api_key,
    base_url=pilot_chat_url,
    timeout=PILOT_TIMEOUT_SECONDS,
)


def quantile_window_pick(
    candidates,
    quantiles,
    prefer_distinct_sources=True,
):
    ordered = sorted(
        candidates,
        key=lambda item: (
            item[
                "raw_text_chars"
            ],
            item[
                "prompt_window_id"
            ],
        ),
    )

    assert ordered

    selected = []
    used_window_ids = set()
    used_source_ids = set()


    for quantile in quantiles:

        target_index = round(
            (
                len(ordered) - 1
            )
            * quantile
        )

        candidate_indices = sorted(
            range(
                len(ordered)
            ),
            key=lambda index: (
                abs(
                    index
                    - target_index
                ),
                index,
            ),
        )

        chosen = None


        if prefer_distinct_sources:

            for index in (
                candidate_indices
            ):

                candidate = (
                    ordered[index]
                )

                if (
                    candidate[
                        "prompt_window_id"
                    ]
                    in used_window_ids
                ):
                    continue

                if (
                    candidate[
                        "source_id"
                    ]
                    in used_source_ids
                ):
                    continue

                chosen = candidate
                break


        if chosen is None:

            for index in (
                candidate_indices
            ):

                candidate = (
                    ordered[index]
                )

                if (
                    candidate[
                        "prompt_window_id"
                    ]
                    in used_window_ids
                ):
                    continue

                chosen = candidate
                break


        assert chosen is not None

        selected.append(
            chosen
        )

        used_window_ids.add(
            chosen[
                "prompt_window_id"
            ]
        )

        used_source_ids.add(
            chosen[
                "source_id"
            ]
        )


    return selected


book_candidates = [
    window
    for window in prompt_windows
    if (
        window[
            "source_type"
        ]
        == "book"
    )
]


transcript_candidates = [
    window
    for window in prompt_windows
    if (
        window[
            "source_type"
        ]
        == "transcript"
    )
]


website_primary_candidates = [
    window
    for window in prompt_windows
    if (
        window[
            "source_type"
        ]
        == "website"
        and window[
            "claim_discovery_role"
        ]
        == "PRIMARY_CLAIM_DISCOVERY"
    )
]


website_secondary_candidates = [
    window
    for window in prompt_windows
    if (
        window[
            "source_type"
        ]
        == "website"
        and window[
            "claim_discovery_role"
        ]
        == (
            "SECONDARY_CLAIM_DISCOVERY_"
            "WITH_ATTRIBUTION"
        )
    )
]


assert len(
    book_candidates
) >= 5

assert len(
    transcript_candidates
) >= 5

assert len(
    website_primary_candidates
) >= 2

assert len(
    website_secondary_candidates
) >= 3


pilot_windows = []


pilot_windows.extend(
    quantile_window_pick(
        book_candidates,
        [
            0.50,
            0.75,
            0.90,
            0.95,
            0.99,
        ],
    )
)


pilot_windows.extend(
    quantile_window_pick(
        transcript_candidates,
        [
            0.50,
            0.75,
            0.90,
            0.95,
            0.99,
        ],
    )
)


pilot_windows.extend(
    quantile_window_pick(
        website_primary_candidates,
        [
            0.50,
            0.90,
        ],
    )
)


pilot_windows.extend(
    quantile_window_pick(
        website_secondary_candidates,
        [
            0.50,
            0.90,
            0.99,
        ],
    )
)


pilot_windows = sorted(
    pilot_windows,
    key=lambda item: (
        item[
            "source_type"
        ],
        item[
            "claim_discovery_role"
        ],
        item[
            "raw_text_chars"
        ],
        item[
            "prompt_window_id"
        ],
    ),
)


assert (
    len(
        pilot_windows
    )
    == 15
)


assert (
    len({
        item[
            "prompt_window_id"
        ]
        for item in (
            pilot_windows
        )
    })
    == 15
)


def materialize_pilot_window(
    window,
):
    slice_map = {}
    prompt_parts = []


    for slice_index, item in enumerate(
        window[
            "slices"
        ],
        1,
    ):

        slice_id = (
            f"S{slice_index:02d}"
        )

        document_id = (
            item[
                "document_id"
            ]
        )

        text = (
            accepted_records_by_id[
                document_id
            ][
                canonical_text_field
            ]
        )

        char_start = (
            item[
                "char_start"
            ]
        )

        char_end = (
            item[
                "char_end"
            ]
        )

        slice_text = text[
            char_start:
            char_end
        ]


        assert (
            len(
                slice_text
            )
            == (
                char_end
                - char_start
            )
        )


        evidence_allowed = (
            not item[
                "is_overlap"
            ]
        )


        slice_record = {
            "slice_id": (
                slice_id
            ),
            "document_id": (
                document_id
            ),
            "char_start": (
                char_start
            ),
            "char_end": (
                char_end
            ),
            "is_overlap": (
                item[
                    "is_overlap"
                ]
            ),
            "evidence_allowed": (
                evidence_allowed
            ),
            "text": (
                slice_text
            ),
        }


        slice_map[
            slice_id
        ] = (
            slice_record
        )


        prompt_parts.append(
            (
                f"<<<SLICE {slice_id} "
                f"document_id={document_id} "
                f"evidence_allowed="
                f"{str(evidence_allowed).lower()} "
                f"char_start={char_start} "
                f"char_end={char_end}>>>\n"
                f"{slice_text}\n"
                f"<<<END SLICE {slice_id}>>>"
            )
        )


    return (
        "\n\n".join(
            prompt_parts
        ),
        slice_map,
    )


def parse_model_json(
    content,
):
    assert isinstance(
        content,
        str,
    )

    cleaned = re.sub(
        r"<think>.*?</think>",
        "",
        content,
        flags=(
            re.IGNORECASE
            | re.DOTALL
        ),
    ).strip()


    start = cleaned.find(
        "{"
    )

    end = cleaned.rfind(
        "}"
    )


    if (
        start < 0
        or end < start
    ):
        raise ValueError(
            "No JSON object found "
            "in model response."
        )


    return json.loads(
        cleaned[
            start:
            end + 1
        ]
    )


def resolve_source_quote(
    slice_text,
    proposed_quote,
):
    if not isinstance(
        proposed_quote,
        str,
    ):
        raise ValueError(
            "Evidence quote is not a string."
        )


    if not proposed_quote.strip():
        raise ValueError(
            "Evidence quote is empty."
        )


    exact_occurrences = [
        match
        for match in re.finditer(
            re.escape(
                proposed_quote
            ),
            slice_text,
        )
    ]


    if (
        len(
            exact_occurrences
        )
        == 1
    ):

        match = (
            exact_occurrences[0]
        )

        return {
            "quote_exact": (
                match.group(0)
            ),
            "relative_start": (
                match.start()
            ),
            "relative_end": (
                match.end()
            ),
            "resolution": (
                "EXACT"
            ),
        }


    if (
        len(
            exact_occurrences
        )
        > 1
    ):
        raise ValueError(
            "Evidence quote must appear "
            "exactly once in its slice. "
            f"Found={len(exact_occurrences)}"
        )


    tokens = (
        proposed_quote.strip().split()
    )


    if not tokens:
        raise ValueError(
            "Evidence quote has no tokens."
        )


    whitespace_pattern = (
        r"\s+".join(
            re.escape(
                token
            )
            for token in tokens
        )
    )


    whitespace_matches = list(
        re.finditer(
            whitespace_pattern,
            slice_text,
        )
    )


    if (
        len(
            whitespace_matches
        )
        == 1
    ):

        match = (
            whitespace_matches[0]
        )

        return {
            "quote_exact": (
                match.group(0)
            ),
            "relative_start": (
                match.start()
            ),
            "relative_end": (
                match.end()
            ),
            "resolution": (
                "WHITESPACE_ONLY_RECOVERED_"
                "FROM_SOURCE"
            ),
        }


    raise ValueError(
        "Evidence quote must appear "
        "exactly once in its slice. "
        f"Found={len(whitespace_matches)}"
    )


def stable_raw_claim_id(
    window_id,
    claim_text,
    evidence_items,
):
    material = stable_json({
        "window_id": (
            window_id
        ),
        "claim_text": (
            claim_text
        ),
        "evidence": sorted(
            [
                {
                    "slice_id": (
                        item[
                            "slice_id"
                        ]
                    ),
                    "quote_exact": (
                        item[
                            "quote_exact"
                        ]
                    ),
                }
                for item in (
                    evidence_items
                )
            ],
            key=lambda item: (
                item[
                    "slice_id"
                ],
                item[
                    "quote_exact"
                ],
            ),
        ),
    })


    digest = hashlib.sha256(
        material.encode(
            "utf-8"
        )
    ).hexdigest()[:32]


    return (
        f"raw-claim:{digest}"
    )


def stable_evidence_id(
    raw_claim_id,
    document_id,
    char_start,
    char_end,
    quote_exact,
):
    material = stable_json({
        "raw_claim_id": (
            raw_claim_id
        ),
        "document_id": (
            document_id
        ),
        "char_start": (
            char_start
        ),
        "char_end": (
            char_end
        ),
        "quote_exact": (
            quote_exact
        ),
    })


    digest = hashlib.sha256(
        material.encode(
            "utf-8"
        )
    ).hexdigest()[:32]


    return (
        f"claim-evidence:{digest}"
    )


def validate_and_build_pilot_bundle(
    window,
    payload,
    slice_map,
):
    if not isinstance(
        payload,
        dict,
    ):
        raise ValueError(
            "Model payload is not an object."
        )


    claims = payload.get(
        "claims"
    )


    if not isinstance(
        claims,
        list,
    ):
        raise ValueError(
            "'claims' must be a list."
        )


    if (
        len(
            claims
        )
        > MAX_CLAIMS_PER_WINDOW
    ):
        raise ValueError(
            "Model returned more than "
            f"{MAX_CLAIMS_PER_WINDOW} claims."
        )


    raw_claims = []
    evidence_links = []

    normalized_claim_texts = set()


    for item in claims:

        if not isinstance(
            item,
            dict,
        ):
            raise ValueError(
                "Claim item is not an object."
            )


        claim_text = str(
            item.get(
                "claim_text_fa",
                "",
            )
        ).strip()


        claim_type = str(
            item.get(
                "claim_type",
                "",
            )
        ).strip()


        evidence_items = item.get(
            "evidence"
        )


        if not claim_text:
            raise ValueError(
                "Empty claim_text_fa."
            )


        if (
            claim_type
            not in ALLOWED_CLAIM_TYPES
        ):
            raise ValueError(
                "Invalid claim_type: "
                f"{claim_type!r}"
            )


        if (
            not isinstance(
                evidence_items,
                list,
            )
            or not evidence_items
        ):
            raise ValueError(
                "Claim has no evidence."
            )


        if (
            len(
                evidence_items
            )
            > MAX_EVIDENCE_ITEMS_PER_CLAIM
        ):
            raise ValueError(
                "Claim contains too many "
                "evidence items."
            )


        normalized_claim_text = (
            " ".join(
                claim_text.split()
            )
            .casefold()
        )


        if (
            normalized_claim_text
            in normalized_claim_texts
        ):
            raise ValueError(
                "Duplicate Claim text "
                "inside the same Window."
            )


        normalized_claim_texts.add(
            normalized_claim_text
        )


        validated_evidence = []
        evidence_span_keys = set()


        for evidence_item in (
            evidence_items
        ):

            if not isinstance(
                evidence_item,
                dict,
            ):
                raise ValueError(
                    "Evidence item is not "
                    "an object."
                )


            slice_id = str(
                evidence_item.get(
                    "slice_id",
                    "",
                )
            ).strip()


            proposed_quote = (
                evidence_item.get(
                    "quote_exact",
                    ""
                )
            )


            if (
                slice_id
                not in slice_map
            ):
                raise ValueError(
                    "Evidence references "
                    "an unknown slice: "
                    f"{slice_id!r}"
                )


            slice_record = (
                slice_map[
                    slice_id
                ]
            )


            if not slice_record[
                "evidence_allowed"
            ]:
                raise ValueError(
                    "Evidence may not use "
                    "an overlap-only slice. "
                    f"Invalid slice={slice_id}"
                )


            resolved_quote = (
                resolve_source_quote(
                    slice_record[
                        "text"
                    ],
                    proposed_quote,
                )
            )


            absolute_start = (
                slice_record[
                    "char_start"
                ]
                + resolved_quote[
                    "relative_start"
                ]
            )


            absolute_end = (
                slice_record[
                    "char_start"
                ]
                + resolved_quote[
                    "relative_end"
                ]
            )


            span_key = (
                slice_record[
                    "document_id"
                ],
                absolute_start,
                absolute_end,
            )


            if (
                span_key
                in evidence_span_keys
            ):
                raise ValueError(
                    "Duplicate evidence span "
                    "inside the same Claim."
                )


            evidence_span_keys.add(
                span_key
            )


            validated_evidence.append({
                "slice_id": (
                    slice_id
                ),
                "quote_exact": (
                    resolved_quote[
                        "quote_exact"
                    ]
                ),
                "document_id": (
                    slice_record[
                        "document_id"
                    ]
                ),
                "char_start": (
                    absolute_start
                ),
                "char_end": (
                    absolute_end
                ),
            })


        raw_claim_id = (
            stable_raw_claim_id(
                window[
                    "prompt_window_id"
                ],
                claim_text,
                validated_evidence,
            )
        )


        source_document_ids = sorted({
            item[
                "document_id"
            ]
            for item in (
                validated_evidence
            )
        })


        raw_claim = {
            "raw_claim_id": (
                raw_claim_id
            ),
            "claim_text_fa": (
                claim_text
            ),
            "claim_type": (
                claim_type
            ),
            "source_document_ids": (
                source_document_ids
            ),
            "authority_class": (
                window[
                    "authority_class"
                ]
            ),
            "claim_discovery_role": (
                window[
                    "claim_discovery_role"
                ]
            ),
            "extraction_unit_id": (
                window[
                    "extraction_unit_id"
                ]
            ),
            "prompt_window_id": (
                window[
                    "prompt_window_id"
                ]
            ),
            "grounding_status": (
                "PASS_EXACT_EVIDENCE"
            ),
            "atomicity_status": (
                "PENDING_SEMANTIC_REVIEW"
            ),
            "validation_status": (
                "PASS_STRUCTURAL_GROUNDING"
            ),
            "extraction_model": (
                pilot_chat_model
            ),
            "extraction_version": (
                PILOT_VERSION
            ),
        }


        raw_claims.append(
            raw_claim
        )


        for evidence in (
            validated_evidence
        ):

            document_record = (
                accepted_records_by_id[
                    evidence[
                        "document_id"
                    ]
                ]
            )


            evidence_id = (
                stable_evidence_id(
                    raw_claim_id,
                    evidence[
                        "document_id"
                    ],
                    evidence[
                        "char_start"
                    ],
                    evidence[
                        "char_end"
                    ],
                    evidence[
                        "quote_exact"
                    ],
                )
            )


            evidence_links.append({
                "evidence_id": (
                    evidence_id
                ),
                "raw_claim_id": (
                    raw_claim_id
                ),
                "document_id": (
                    evidence[
                        "document_id"
                    ]
                ),
                "source_id": (
                    document_record.get(
                        "source_id"
                    )
                ),
                "source_type": (
                    document_record.get(
                        "source_type"
                    )
                ),
                "authority_class": (
                    window[
                        "authority_class"
                    ]
                ),
                "claim_discovery_role": (
                    window[
                        "claim_discovery_role"
                    ]
                ),
                "evidence_quote_exact": (
                    evidence[
                        "quote_exact"
                    ]
                ),
                "locator": {
                    "source_locator": (
                        document_record.get(
                            "locator"
                        )
                    ),
                    "text_char_start": (
                        evidence[
                            "char_start"
                        ]
                    ),
                    "text_char_end": (
                        evidence[
                            "char_end"
                        ]
                    ),
                },
                "attribution_snapshot": (
                    document_record.get(
                        "attribution"
                    )
                ),
                "evidence_text_sha256": (
                    hashlib.sha256(
                        evidence[
                            "quote_exact"
                        ].encode(
                            "utf-8"
                        )
                    ).hexdigest()
                ),
                "prompt_window_id": (
                    window[
                        "prompt_window_id"
                    ]
                ),
                "extraction_version": (
                    PILOT_VERSION
                ),
            })


    return {
        "prompt_window_id": (
            window[
                "prompt_window_id"
            ]
        ),
        "raw_claims": (
            raw_claims
        ),
        "evidence_links": (
            evidence_links
        ),
    }


def build_claim_prompt(
    window,
    context_text,
    slice_map,
    retry_feedback=None,
):
    eligible_slice_ids = [
        slice_id
        for slice_id, item in (
            slice_map.items()
        )
        if item[
            "evidence_allowed"
        ]
    ]


    system_message = """
You extract atomic, source-grounded knowledge claims from Persian financial-education sources.

Use only the supplied source slices.
Do not use outside knowledge.
Do not create questions or answers.
Return one valid JSON object only.
Do not return markdown, analysis, explanations, or code fences.

A claim must be a useful standalone knowledge statement.
Prefer atomic claims over compound claims.
If no useful independent claim exists, return {"claims": []}.

Every claim must have direct source evidence.
Only slices explicitly listed as eligible evidence may be cited.
A quote must be copied from the source, not paraphrased.
""".strip()


    feedback_block = ""


    if retry_feedback:

        feedback_block = f"""

IMPORTANT RETRY CORRECTION:

Your previous response failed deterministic validation.

Validator error:
{retry_feedback}

Return the complete JSON again after correcting the error.

Correction rules:
- Return syntactically valid JSON.
- Use ONLY an allowed claim_type.
- Use ONLY eligible evidence slice_ids.
- Never cite evidence_allowed=false slices.
- Copy quote_exact directly from the cited slice.
- Preserve Persian letters, punctuation and wording.
- If an exact long quote is difficult, use a shorter distinctive verbatim quote that still directly supports the Claim.
- The quote must occur exactly once in that slice.
"""


    user_message = f"""
Extract at most {MAX_CLAIMS_PER_WINDOW} independent atomic claims.

Allowed claim_type values:
{json.dumps(sorted(ALLOWED_CLAIM_TYPES), ensure_ascii=False)}

Eligible evidence slice_ids:
{json.dumps(eligible_slice_ids, ensure_ascii=False)}

Required JSON schema:

{{
  "claims": [
    {{
      "claim_text_fa": "یک گزاره مستقل و دقیق به فارسی",
      "claim_type": "ONE_ALLOWED_TYPE",
      "evidence": [
        {{
          "slice_id": "S01",
          "quote_exact": "متن دقیق از همان Slice"
        }}
      ]
    }}
  ]
}}

Rules:
1. Use only the supplied source.
2. Do not infer missing facts.
3. Do not combine unrelated facts.
4. Do not create questions.
5. Do not create answers.
6. Evidence must directly support the Claim.
7. Evidence slice_id MUST be one of:
   {json.dumps(eligible_slice_ids, ensure_ascii=False)}
8. Slices with evidence_allowed=false are CONTEXT ONLY.
9. Never cite an overlap/context-only slice as Evidence.
10. quote_exact must be copied directly from the selected eligible slice.
11. Do not insert ellipses or rewrite the Evidence quote.
12. Prefer a short but sufficient distinctive exact quote.
13. Do not repeat one Claim using different wording.
14. If no useful independent Claim exists, return an empty claims list.

Window metadata:
source_type={window['source_type']}
authority_class={window['authority_class']}
claim_discovery_role={window['claim_discovery_role']}

SOURCE SLICES:

{context_text}

{feedback_block}
""".strip()


    return (
        system_message,
        user_message,
    )


def call_pilot_model(
    system_message,
    user_message,
):
    response = (
        pilot_client
        .chat
        .completions
        .create(
            model=(
                pilot_chat_model
            ),
            messages=[
                {
                    "role": (
                        "system"
                    ),
                    "content": (
                        system_message
                    ),
                },
                {
                    "role": (
                        "user"
                    ),
                    "content": (
                        user_message
                    ),
                },
            ],
            temperature=(
                PILOT_TEMPERATURE
            ),
            max_tokens=(
                PILOT_MAX_TOKENS
            ),
        )
    )


    content = (
        response
        .choices[0]
        .message
        .content
    )


    if not isinstance(
        content,
        str,
    ):
        raise ValueError(
            "Model response content "
            "is not a string."
        )


    return content


pilot_cache_root.mkdir(
    parents=True,
    exist_ok=True,
)


pilot_bundles = []
pilot_failures = []

pilot_api_calls_this_run = 0
pilot_cache_hits_this_run = 0


print(
    "=== NB12i REAL CLAIM EXTRACTION PILOT ==="
)

print()
print(
    "Reference model:",
    reference_model,
)

print(
    "Configured model:",
    pilot_chat_model,
)

print(
    "Pilot windows:",
    len(
        pilot_windows
    ),
)

print()


for index, window in enumerate(
    pilot_windows,
    1,
):

    cache_path = (
        pilot_cache_root
        / (
            window[
                "prompt_window_id"
            ].replace(
                ":",
                "_",
            )
            + ".json"
        )
    )


    print(
        f"[{index:02d}/{len(pilot_windows):02d}]",
        window[
            "source_type"
        ],
        "| role:",
        window[
            "claim_discovery_role"
        ],
        "| chars:",
        window[
            "raw_text_chars"
        ],
        "|",
        window[
            "prompt_window_id"
        ],
    )


    if cache_path.exists():

        cached_bundle = load_json(
            cache_path
        )


        assert (
            cached_bundle.get(
                "prompt_window_id"
            )
            == window[
                "prompt_window_id"
            ]
        )


        assert isinstance(
            cached_bundle.get(
                "raw_claims"
            ),
            list,
        )


        assert isinstance(
            cached_bundle.get(
                "evidence_links"
            ),
            list,
        )


        pilot_bundles.append(
            cached_bundle
        )

        pilot_cache_hits_this_run += 1


        print(
            "   cache: HIT",
            "| claims:",
            len(
                cached_bundle[
                    "raw_claims"
                ]
            ),
        )

        continue


    (
        context_text,
        slice_map,
    ) = materialize_pilot_window(
        window
    )


    validated_bundle = None
    last_error = None


    for attempt in range(
        1,
        PILOT_MAX_ATTEMPTS + 1,
    ):

        (
            system_message,
            user_message,
        ) = build_claim_prompt(
            window,
            context_text,
            slice_map,
            retry_feedback=(
                last_error
            ),
        )


        try:

            content = call_pilot_model(
                system_message,
                user_message,
            )

            pilot_api_calls_this_run += 1


            payload = (
                parse_model_json(
                    content
                )
            )


            validated_bundle = (
                validate_and_build_pilot_bundle(
                    window,
                    payload,
                    slice_map,
                )
            )


            break


        except Exception as exc:

            last_error = (
                f"{type(exc).__name__}: "
                f"{exc}"
            )


            print(
                f"   attempt {attempt}:",
                "INVALID",
                "|",
                last_error,
            )


            if (
                attempt
                < PILOT_MAX_ATTEMPTS
            ):
                time.sleep(
                    min(
                        2.0,
                        float(
                            attempt
                        ),
                    )
                )


    if validated_bundle is None:

        pilot_failures.append({
            "prompt_window_id": (
                window[
                    "prompt_window_id"
                ]
            ),
            "source_type": (
                window[
                    "source_type"
                ]
            ),
            "last_error": (
                last_error
            ),
        })


        print(
            "   result: FAIL"
        )

        continue


    cache_text = (
        json.dumps(
            validated_bundle,
            ensure_ascii=False,
            indent=2,
            sort_keys=True,
        )
        + "\n"
    )


    with cache_path.open(
        "x",
        encoding="utf-8",
    ) as file:
        file.write(
            cache_text
        )


    pilot_bundles.append(
        validated_bundle
    )


    print(
        "   result: VALID",
        "| claims:",
        len(
            validated_bundle[
                "raw_claims"
            ]
        ),
    )


print()


assert not pilot_failures, (
    "Pilot has unresolved Window failures:\n"
    + json.dumps(
        pilot_failures,
        ensure_ascii=False,
        indent=2,
    )
)


assert (
    len(
        pilot_bundles
    )
    == len(
        pilot_windows
    )
)


pilot_raw_claims = sorted(
    [
        claim
        for bundle in (
            pilot_bundles
        )
        for claim in (
            bundle[
                "raw_claims"
            ]
        )
    ],
    key=lambda item: (
        item[
            "raw_claim_id"
        ]
    ),
)


pilot_evidence_links = sorted(
    [
        evidence
        for bundle in (
            pilot_bundles
        )
        for evidence in (
            bundle[
                "evidence_links"
            ]
        )
    ],
    key=lambda item: (
        item[
            "evidence_id"
        ]
    ),
)


raw_claim_ids = [
    item[
        "raw_claim_id"
    ]
    for item in (
        pilot_raw_claims
    )
]


evidence_ids = [
    item[
        "evidence_id"
    ]
    for item in (
        pilot_evidence_links
    )
]


assert (
    len(
        raw_claim_ids
    )
    == len(
        set(
            raw_claim_ids
        )
    )
), (
    "Duplicate raw_claim_id detected."
)


assert (
    len(
        evidence_ids
    )
    == len(
        set(
            evidence_ids
        )
    )
), (
    "Duplicate evidence_id detected."
)


claim_count_by_source = Counter()
claim_count_by_type = Counter()
claim_count_by_role = Counter()


window_by_id = {
    item[
        "prompt_window_id"
    ]: item
    for item in (
        pilot_windows
    )
}


for claim in (
    pilot_raw_claims
):

    window = (
        window_by_id[
            claim[
                "prompt_window_id"
            ]
        ]
    )


    claim_count_by_source[
        window[
            "source_type"
        ]
    ] += 1


    claim_count_by_type[
        claim[
            "claim_type"
        ]
    ] += 1


    claim_count_by_role[
        claim[
            "claim_discovery_role"
        ]
    ] += 1


for source_type in [
    "book",
    "transcript",
    "website",
]:

    assert (
        claim_count_by_source[
            source_type
        ]
        > 0
    ), (
        "Pilot produced no Claims for "
        f"source_type={source_type!r}."
    )


pilot_window_ids = [
    window[
        "prompt_window_id"
    ]
    for window in (
        pilot_windows
    )
]


pilot_raw_text = "".join(
    json.dumps(
        item,
        ensure_ascii=False,
        sort_keys=True,
    )
    + "\n"
    for item in (
        pilot_raw_claims
    )
)


pilot_evidence_text = "".join(
    json.dumps(
        item,
        ensure_ascii=False,
        sort_keys=True,
    )
    + "\n"
    for item in (
        pilot_evidence_links
    )
)


pilot_raw_sha256 = (
    hashlib.sha256(
        pilot_raw_text.encode(
            "utf-8"
        )
    ).hexdigest()
)


pilot_evidence_sha256 = (
    hashlib.sha256(
        pilot_evidence_text.encode(
            "utf-8"
        )
    ).hexdigest()
)


pilot_report = {
    "report_version": (
        "claim_extraction_pilot_report_v1"
    ),

    "pilot_version": (
        PILOT_VERSION
    ),

    "reference_model": (
        reference_model
    ),

    "configured_model": (
        pilot_chat_model
    ),

    "generation_contract": {
        "temperature": (
            PILOT_TEMPERATURE
        ),
        "max_tokens": (
            PILOT_MAX_TOKENS
        ),
        "timeout_seconds": (
            PILOT_TIMEOUT_SECONDS
        ),
        "max_attempts_per_window": (
            PILOT_MAX_ATTEMPTS
        ),
        "adaptive_retry": True,
        "whitespace_only_source_quote_recovery": True,
    },

    "selection_contract": {
        "total_windows": len(
            pilot_windows
        ),
        "window_ids": (
            pilot_window_ids
        ),
        "book_windows": 5,
        "transcript_windows": 5,
        "website_windows": 5,
        "website_primary_windows": 2,
        "website_secondary_windows": 3,
    },

    "results": {
        "raw_claims": len(
            pilot_raw_claims
        ),
        "evidence_links": len(
            pilot_evidence_links
        ),
        "claims_by_source_type": dict(
            sorted(
                claim_count_by_source.items()
            )
        ),
        "claims_by_type": dict(
            sorted(
                claim_count_by_type.items()
            )
        ),
        "claims_by_role": dict(
            sorted(
                claim_count_by_role.items()
            )
        ),
        "unresolved_window_failures": 0,
        "duplicate_raw_claim_ids": 0,
        "duplicate_evidence_ids": 0,
        "structural_grounding_failures": 0,
    },

    "semantic_status": (
        "NOT_YET_RUN"
    ),

    "full_registry_status": (
        "NOT_STARTED"
    ),

    "artifacts": {
        "raw_claims": (
            pilot_raw_path.name
        ),
        "raw_claims_sha256": (
            pilot_raw_sha256
        ),
        "evidence_links": (
            pilot_evidence_path.name
        ),
        "evidence_links_sha256": (
            pilot_evidence_sha256
        ),
    },
}


pilot_report_text = (
    json.dumps(
        pilot_report,
        ensure_ascii=False,
        indent=2,
        sort_keys=True,
    )
    + "\n"
)


def publish_no_overwrite(
    path,
    content,
):
    if path.exists():

        existing = (
            path.read_text(
                encoding="utf-8"
            )
        )

        assert (
            existing
            == content
        ), (
            "Existing Pilot Artifact differs "
            f"from staged output: {path.name}. "
            "Refusing overwrite."
        )

        return (
            "REUSED_EXISTING_IDENTICAL"
        )


    with path.open(
        "x",
        encoding="utf-8",
    ) as file:
        file.write(
            content
        )


    return (
        "CREATED_NEW"
    )


pilot_raw_write_status = (
    publish_no_overwrite(
        pilot_raw_path,
        pilot_raw_text,
    )
)


pilot_evidence_write_status = (
    publish_no_overwrite(
        pilot_evidence_path,
        pilot_evidence_text,
    )
)


pilot_report_write_status = (
    publish_no_overwrite(
        pilot_report_path,
        pilot_report_text,
    )
)


disk_pilot_raw_claims = (
    load_jsonl(
        pilot_raw_path
    )
)

disk_pilot_evidence_links = (
    load_jsonl(
        pilot_evidence_path
    )
)

disk_pilot_report = (
    load_json(
        pilot_report_path
    )
)


assert (
    disk_pilot_raw_claims
    == pilot_raw_claims
)

assert (
    disk_pilot_evidence_links
    == pilot_evidence_links
)

assert (
    disk_pilot_report
    == pilot_report
)


assert (
    file_sha256(
        pilot_raw_path
    )
    == pilot_raw_sha256
)

assert (
    file_sha256(
        pilot_evidence_path
    )
    == pilot_evidence_sha256
)


pilot_claim_extraction_status = (
    "PASS"
)


print(
    "=== NB12i CLAIM EXTRACTION PILOT RESULT ==="
)

print()
print(
    "Pilot Prompt Windows:",
    len(
        pilot_windows
    ),
)

print(
    "API calls this run:",
    pilot_api_calls_this_run,
)

print(
    "Validated cache hits this run:",
    pilot_cache_hits_this_run,
)


print()
print(
    "Raw Claims:",
    len(
        pilot_raw_claims
    ),
)

print(
    "Evidence Links:",
    len(
        pilot_evidence_links
    ),
)


print()
print(
    "Claims by source type:",
    dict(
        sorted(
            claim_count_by_source.items()
        )
    ),
)

print(
    "Claims by authority role:",
    dict(
        sorted(
            claim_count_by_role.items()
        )
    ),
)

print(
    "Claims by type:",
    dict(
        sorted(
            claim_count_by_type.items()
        )
    ),
)


print()
print(
    "Unresolved Window failures:",
    len(
        pilot_failures
    ),
)

print(
    "Duplicate Raw Claim IDs:",
    (
        len(
            raw_claim_ids
        )
        - len(
            set(
                raw_claim_ids
            )
        )
    ),
)

print(
    "Duplicate Evidence IDs:",
    (
        len(
            evidence_ids
        )
        - len(
            set(
                evidence_ids
            )
        )
    ),
)

print(
    "Structural grounding failures:",
    0,
)


print()
print(
    "Raw Claim Artifact:",
    pilot_raw_path.relative_to(
        project_root
    ),
)

print(
    "Raw Claim SHA-256:",
    pilot_raw_sha256,
)

print(
    "Raw Claim write status:",
    pilot_raw_write_status,
)


print()
print(
    "Evidence Artifact:",
    pilot_evidence_path.relative_to(
        project_root
    ),
)

print(
    "Evidence SHA-256:",
    pilot_evidence_sha256,
)

print(
    "Evidence write status:",
    pilot_evidence_write_status,
)


print()
print(
    "Pilot Report:",
    pilot_report_path.relative_to(
        project_root
    ),
)

print(
    "Pilot Report write status:",
    pilot_report_write_status,
)


print()
print(
    "=== SAMPLE REAL CLAIMS ==="
)


evidence_by_raw_claim_id = defaultdict(
    list
)


for evidence in (
    pilot_evidence_links
):

    evidence_by_raw_claim_id[
        evidence[
            "raw_claim_id"
        ]
    ].append(
        evidence
    )


for claim in (
    pilot_raw_claims[:12]
):

    claim_evidence = (
        evidence_by_raw_claim_id[
            claim[
                "raw_claim_id"
            ]
        ]
    )


    source_type = (
        claim_evidence[0][
            "source_type"
        ]
        if claim_evidence
        else "UNKNOWN"
    )


    print()
    print(
        "Raw Claim ID:",
        claim[
            "raw_claim_id"
        ],
    )

    print(
        "Source type:",
        source_type,
    )

    print(
        "Claim type:",
        claim[
            "claim_type"
        ],
    )

    print(
        "Claim:",
        claim[
            "claim_text_fa"
        ],
    )


    for evidence in (
        claim_evidence
    ):

        quote = (
            evidence[
                "evidence_quote_exact"
            ]
        )


        if len(
            quote
        ) > 220:

            quote = (
                quote[:217]
                + "..."
            )


        print(
            "Evidence:",
            quote,
        )


print()
print(
    "Semantic Quality Gate:",
    "NOT_YET_RUN",
)

print(
    "Full Registry Extraction:",
    "NOT_STARTED",
)


print()
print(
    "CODE CELL 10 GATE:",
    pilot_claim_extraction_status,
)
```

<h2>Semantic Quality Gate برای Claimهای واقعی Pilot</h2>

<p>
Pilot مرحلهٔ قبل از نظر ساختاری و Grounding کامل PASS شد:
تمام ۱۵ Prompt Window پردازش شدند و ۵۹ Raw Claim با ۶۲ Evidence Link معتبر ساخته شدند.
</p>

<p>
با این حال، Exact Evidence به‌تنهایی تضمین نمی‌کند که یک Claim برای Claim Registry مناسب باشد.
ممکن است Claim از متن واقعی آمده باشد اما همچنان ناقص، وابسته به Context، بیش از حد کلی، غیراتمی یا صرفاً مربوط به Metadata متن باشد.
</p>

<h3>هدف این مرحله</h3>

<p>
هر یک از ۵۹ Claim واقعی Pilot از نظر کیفیت معنایی مستقل adjudicate می‌شود.
در این مرحله Claim جدیدی از Corpus استخراج نمی‌شود و Full Registry نیز هنوز اجرا نمی‌شود.
</p>

<h3>شش معیار Semantic Quality</h3>

<ul>
<li><strong>Entailment:</strong> آیا Claim واقعاً از Evidence نتیجه می‌شود یا چیزی بیش از متن اضافه کرده است؟</li>
<li><strong>Atomicity:</strong> آیا Claim یک واحد دانشی مستقل است یا چند گزاره را با هم ترکیب کرده است؟</li>
<li><strong>Standalone:</strong> آیا Claim بدون نیاز به عباراتی مانند «این»، «آن»، «او»، «اینجا» یا Context گمشده قابل فهم است؟</li>
<li><strong>Evidence Sufficiency:</strong> آیا Evidence برای اثبات کل Claim کافی است یا فقط بخشی از آن را نشان می‌دهد؟</li>
<li><strong>Domain Knowledge:</strong> آیا Claim واقعاً دانش قابل استفاده برای Registry است یا صرفاً Metadata، توضیح ویراستاری، Fragment یا متن اداری است؟</li>
<li><strong>Claim Type Fit:</strong> آیا Claim Type انتخاب‌شده با ماهیت گزاره سازگار است؟</li>
</ul>

<h3>سه تصمیم ممکن</h3>

<p><strong>ACCEPT</strong></p>

<p>
Claim بدون تغییر برای ادامهٔ Pipeline مناسب است و تمام معیارهای اصلی را PASS می‌کند.
</p>

<p><strong>REPAIR</strong></p>

<p>
دانش و Evidence معتبر هستند، اما متن Claim یا Claim Type نیاز به اصلاح محدود دارد.
Repair فقط زمانی مجاز است که بدون افزودن Fact جدید و فقط با استفاده از همان Evidence قابل انجام باشد.
</p>

<p><strong>REJECT</strong></p>

<p>
Claim برای Registry مناسب نیست؛ مثلاً Evidence ناقص است، Claim واقعاً از متن نتیجه نمی‌شود، گزاره صرفاً Metadata منبع است یا بدون Context گمشده قابل بازیابی نیست.
</p>

<h3>Issueهای کنترل‌شده</h3>

<ul>
<li><code>UNSUPPORTED_OR_OVERSTATED</code></li>
<li><code>NOT_ATOMIC</code></li>
<li><code>CONTEXT_DEPENDENT</code></li>
<li><code>EVIDENCE_INCOMPLETE</code></li>
<li><code>SOURCE_METADATA_NOT_DOMAIN_KNOWLEDGE</code></li>
<li><code>CLAIM_TYPE_MISMATCH</code></li>
<li><code>TOO_VAGUE</code></li>
<li><code>IMPERATIVE_OR_FRAGMENT</code></li>
<li><code>OTHER</code></li>
</ul>

<h3>Context مورد استفاده برای Adjudication</h3>

<p>
برای اینکه Judge فقط یک Quote بریده‌شده را نبیند، علاوه بر Claim و Exact Evidence، مقدار محدودی Context واقعی قبل و بعد از Evidence نیز از Accepted Corpus Materialize می‌شود.
</p>

<p>
این Context فقط برای داوری استفاده می‌شود و هیچ Evidence جدیدی به Claim اضافه نمی‌کند.
</p>

<h3>Batch و Cache</h3>

<p>
Claimها در Batchهای کوچک چهار‌تایی adjudicate می‌شوند تا Prompt قابل کنترل باقی بماند.
هر Batch فقط پس از Parse و Validation کامل در Cache ثبت خواهد شد.
Retry نیز Adaptive است و خطای Validator را به مدل بازمی‌گرداند.
</p>

<h3>Artifactهای این مرحله</h3>

<ul>
<li><code>claim_extraction_pilot_semantic_adjudication_v1.jsonl</code></li>
<li><code>claim_extraction_pilot_semantic_report_v1.json</code></li>
<li>Cache مستقل Semantic Adjudication.</li>
</ul>

<h3>اصل مهم</h3>

<p>
PASS شدن Code Cell این مرحله فقط به معنی کامل و معتبر بودن فرآیند Adjudication است.
این Cell به‌صورت خودکار دربارهٔ آماده‌بودن Full Corpus تصمیم نمی‌گیرد.
پس از مشاهدهٔ نسبت ACCEPT / REPAIR / REJECT و نمونه‌های واقعی، Prompt Extraction مرحلهٔ بعد Freeze یا Repair خواهد شد.
</p>

<h3>معیار PASS</h3>

<ul>
<li>تمام ۵۹ Claim دقیقاً یک Semantic Judgment داشته باشند؛</li>
<li>هیچ Claim گمشده یا اضافه وجود نداشته باشد؛</li>
<li>Decision و Checkها با Contract سازگار باشند؛</li>
<li>REPAIR فقط با Evidence موجود انجام شود؛</li>
<li>تمام Issueها از Enum کنترل‌شده باشند؛</li>
<li>هیچ API Failure حل‌نشده باقی نماند؛</li>
<li>Artifactها بدون Overwrite ناامن منتشر شوند؛</li>
<li>Artifactها دوباره از Disk Validate شوند؛</li>
<li>Full Corpus Extraction همچنان شروع نشود.</li>
</ul>

```python
# Code Cell 11

assert (
    pilot_claim_extraction_status
    == "PASS"
), (
    "Code Cell 10 must PASS before "
    "running Semantic Quality Adjudication."
)


SEMANTIC_PROMPT_VERSION = (
    "claim_semantic_adjudication_v1"
)

SEMANTIC_BATCH_SIZE = 4
SEMANTIC_MAX_TOKENS = 3500
SEMANTIC_TEMPERATURE = 0
SEMANTIC_MAX_ATTEMPTS = 3
SEMANTIC_CONTEXT_CHARS = 350


SEMANTIC_CHECK_NAMES = {
    "entailment",
    "atomicity",
    "standalone",
    "evidence_sufficiency",
    "domain_knowledge",
    "claim_type_fit",
}


SEMANTIC_CHECK_VALUES = {
    "PASS",
    "FAIL",
}


SEMANTIC_DECISIONS = {
    "ACCEPT",
    "REPAIR",
    "REJECT",
}


SEMANTIC_ISSUES = {
    "UNSUPPORTED_OR_OVERSTATED",
    "NOT_ATOMIC",
    "CONTEXT_DEPENDENT",
    "EVIDENCE_INCOMPLETE",
    "SOURCE_METADATA_NOT_DOMAIN_KNOWLEDGE",
    "CLAIM_TYPE_MISMATCH",
    "TOO_VAGUE",
    "IMPERATIVE_OR_FRAGMENT",
    "OTHER",
}


semantic_adjudication_path = (
    registry_artifact_root
    / "claim_extraction_pilot_semantic_adjudication_v1.jsonl"
)

semantic_report_path = (
    registry_artifact_root
    / "claim_extraction_pilot_semantic_report_v1.json"
)

semantic_cache_root = (
    registry_artifact_root
    / "claim_extraction_pilot_semantic_cache_v1"
)


# ---------------------------------------------------------
# Reload Pilot artifacts independently from disk.
# ---------------------------------------------------------

semantic_source_claims = load_jsonl(
    pilot_raw_path
)

semantic_source_evidence = load_jsonl(
    pilot_evidence_path
)

semantic_source_report = load_json(
    pilot_report_path
)


assert (
    len(
        semantic_source_claims
    )
    == 59
), (
    "Pilot Raw Claim count changed "
    "before Semantic Adjudication."
)


assert (
    len(
        semantic_source_evidence
    )
    == 62
), (
    "Pilot Evidence count changed "
    "before Semantic Adjudication."
)


assert (
    semantic_source_report[
        "results"
    ][
        "raw_claims"
    ]
    == len(
        semantic_source_claims
    )
)


semantic_claim_by_id = {
    claim[
        "raw_claim_id"
    ]: claim
    for claim in (
        semantic_source_claims
    )
}


assert (
    len(
        semantic_claim_by_id
    )
    == len(
        semantic_source_claims
    )
), (
    "Duplicate Raw Claim IDs detected "
    "while reloading Pilot artifacts."
)


semantic_evidence_by_claim = defaultdict(
    list
)


for evidence in (
    semantic_source_evidence
):

    raw_claim_id = evidence[
        "raw_claim_id"
    ]

    assert (
        raw_claim_id
        in semantic_claim_by_id
    ), (
        "Evidence references an unknown "
        f"Raw Claim: {raw_claim_id}"
    )

    semantic_evidence_by_claim[
        raw_claim_id
    ].append(
        evidence
    )


for raw_claim_id in (
    semantic_claim_by_id
):

    assert (
        semantic_evidence_by_claim[
            raw_claim_id
        ]
    ), (
        "Raw Claim has no Evidence Link: "
        f"{raw_claim_id}"
    )


# ---------------------------------------------------------
# Build compact real source context around each Evidence.
# ---------------------------------------------------------

def materialize_semantic_context(
    evidence,
):
    document_id = evidence[
        "document_id"
    ]

    document_record = (
        accepted_records_by_id[
            document_id
        ]
    )

    source_text = (
        document_record[
            canonical_text_field
        ]
    )

    locator = evidence[
        "locator"
    ]

    evidence_start = locator[
        "text_char_start"
    ]

    evidence_end = locator[
        "text_char_end"
    ]

    quote = evidence[
        "evidence_quote_exact"
    ]


    assert (
        source_text[
            evidence_start:
            evidence_end
        ]
        == quote
    ), (
        "Pilot Evidence no longer matches "
        "Accepted Corpus exactly."
    )


    context_start = max(
        0,
        evidence_start
        - SEMANTIC_CONTEXT_CHARS,
    )

    context_end = min(
        len(
            source_text
        ),
        evidence_end
        + SEMANTIC_CONTEXT_CHARS,
    )


    context_text = (
        source_text[
            context_start:
            context_end
        ]
    )


    return {
        "document_id": (
            document_id
        ),
        "source_type": (
            evidence[
                "source_type"
            ]
        ),
        "source_id": (
            evidence[
                "source_id"
            ]
        ),
        "authority_class": (
            evidence[
                "authority_class"
            ]
        ),
        "claim_discovery_role": (
            evidence[
                "claim_discovery_role"
            ]
        ),
        "quote_exact": (
            quote
        ),
        "context_before_chars": (
            evidence_start
            - context_start
        ),
        "context_after_chars": (
            context_end
            - evidence_end
        ),
        "context_text": (
            context_text
        ),
    }


semantic_items = []


for claim in sorted(
    semantic_source_claims,
    key=lambda item: (
        item[
            "raw_claim_id"
        ]
    ),
):

    raw_claim_id = claim[
        "raw_claim_id"
    ]

    evidence_contexts = [
        materialize_semantic_context(
            evidence
        )
        for evidence in (
            semantic_evidence_by_claim[
                raw_claim_id
            ]
        )
    ]


    semantic_items.append({
        "raw_claim_id": (
            raw_claim_id
        ),
        "claim_text_fa": (
            claim[
                "claim_text_fa"
            ]
        ),
        "claim_type": (
            claim[
                "claim_type"
            ]
        ),
        "authority_class": (
            claim[
                "authority_class"
            ]
        ),
        "claim_discovery_role": (
            claim[
                "claim_discovery_role"
            ]
        ),
        "evidence_contexts": (
            evidence_contexts
        ),
    })


assert (
    len(
        semantic_items
    )
    == 59
)


# ---------------------------------------------------------
# Deterministic semantic batches.
# ---------------------------------------------------------

semantic_batches = [
    semantic_items[
        start:
        start + SEMANTIC_BATCH_SIZE
    ]
    for start in range(
        0,
        len(
            semantic_items
        ),
        SEMANTIC_BATCH_SIZE,
    )
]


def stable_semantic_batch_id(
    batch_items,
):
    material = stable_json({
        "prompt_version": (
            SEMANTIC_PROMPT_VERSION
        ),
        "model": (
            pilot_chat_model
        ),
        "raw_claim_ids": [
            item[
                "raw_claim_id"
            ]
            for item in (
                batch_items
            )
        ],
    })


    digest = hashlib.sha256(
        material.encode(
            "utf-8"
        )
    ).hexdigest()[:24]


    return (
        f"semantic-batch:{digest}"
    )


# ---------------------------------------------------------
# Semantic prompt.
# ---------------------------------------------------------

def build_semantic_prompt(
    batch_items,
    retry_feedback=None,
):
    compact_items = []


    for item in batch_items:

        compact_items.append({
            "raw_claim_id": (
                item[
                    "raw_claim_id"
                ]
            ),
            "claim_text_fa": (
                item[
                    "claim_text_fa"
                ]
            ),
            "claim_type": (
                item[
                    "claim_type"
                ]
            ),
            "authority_class": (
                item[
                    "authority_class"
                ]
            ),
            "claim_discovery_role": (
                item[
                    "claim_discovery_role"
                ]
            ),
            "evidence": [
                {
                    "document_id": (
                        evidence[
                            "document_id"
                        ]
                    ),
                    "quote_exact": (
                        evidence[
                            "quote_exact"
                        ]
                    ),
                    "surrounding_context": (
                        evidence[
                            "context_text"
                        ]
                    ),
                }
                for evidence in (
                    item[
                        "evidence_contexts"
                    ]
                )
            ],
        })


    retry_block = ""


    if retry_feedback:

        retry_block = f"""

Your previous response failed deterministic validation.

Validator error:
{retry_feedback}

Return the COMPLETE JSON response again for every raw_claim_id.
Do not omit any item.
"""


    system_message = """
You are a strict semantic adjudicator for an evidence-grounded Persian financial knowledge registry.

You are NOT extracting new claims.
You are judging claims that were already extracted.

Use only:
1. the claim,
2. its exact evidence,
3. the supplied local source context.

Do not use outside knowledge.

Judge conservatively.

Decision rules:

ACCEPT:
- the full claim is directly supported,
- the claim is atomic enough,
- the claim is understandable as a standalone knowledge statement,
- evidence is sufficient,
- the claim is useful domain knowledge rather than source/editorial metadata,
- claim_type is appropriate.

REPAIR:
- the underlying knowledge and evidence are valid,
- but wording, standalone form, atomicity, vagueness or claim_type can be safely fixed using ONLY the supplied evidence/context,
- no new fact may be introduced.

REJECT:
- evidence does not support the full claim,
- evidence is incomplete,
- the statement is merely source/editorial metadata,
- the statement is only an imperative/fragment,
- required context is missing and cannot be repaired safely,
- or it is otherwise unsuitable as an independent registry claim.

For REPAIR, repaired_claim_text_fa must be a complete standalone Persian claim supported by the supplied evidence.
Do not improve or enrich it with outside knowledge.

Return valid JSON only.
No markdown.
No analysis outside JSON.
""".strip()


    user_message = f"""
Adjudicate every Claim below.

Allowed decisions:
{json.dumps(sorted(SEMANTIC_DECISIONS), ensure_ascii=False)}

Allowed check values:
{json.dumps(sorted(SEMANTIC_CHECK_VALUES), ensure_ascii=False)}

Required checks:
{json.dumps(sorted(SEMANTIC_CHECK_NAMES), ensure_ascii=False)}

Allowed issue codes:
{json.dumps(sorted(SEMANTIC_ISSUES), ensure_ascii=False)}

Allowed claim types:
{json.dumps(sorted(ALLOWED_CLAIM_TYPES), ensure_ascii=False)}

Required output schema:

{{
  "judgments": [
    {{
      "raw_claim_id": "raw-claim:...",
      "decision": "ACCEPT | REPAIR | REJECT",
      "checks": {{
        "entailment": "PASS | FAIL",
        "atomicity": "PASS | FAIL",
        "standalone": "PASS | FAIL",
        "evidence_sufficiency": "PASS | FAIL",
        "domain_knowledge": "PASS | FAIL",
        "claim_type_fit": "PASS | FAIL"
      }},
      "issues": [
        "CONTROLLED_ISSUE_CODE"
      ],
      "repaired_claim_text_fa": null,
      "recommended_claim_type": "ALLOWED_TYPE_OR_NULL",
      "brief_reason_fa": "دلیل کوتاه و مشخص"
    }}
  ]
}}

Consistency rules:

1. ACCEPT requires every check to PASS.
2. ACCEPT must have repaired_claim_text_fa = null.
3. ACCEPT must keep recommended_claim_type equal to the original claim_type.
4. REPAIR requires entailment = PASS.
5. REPAIR requires evidence_sufficiency = PASS.
6. REPAIR requires domain_knowledge = PASS.
7. REPAIR requires a non-empty repaired_claim_text_fa.
8. REPAIR requires recommended_claim_type to be an allowed claim type.
9. REJECT may have repaired_claim_text_fa = null and recommended_claim_type = null.
10. SOURCE_METADATA_NOT_DOMAIN_KNOWLEDGE implies domain_knowledge = FAIL.
11. EVIDENCE_INCOMPLETE implies evidence_sufficiency = FAIL.
12. UNSUPPORTED_OR_OVERSTATED implies entailment = FAIL.
13. Every input raw_claim_id must appear exactly once.
14. Do not add extra raw_claim_ids.

Claims:

{json.dumps(compact_items, ensure_ascii=False, indent=2)}

{retry_block}
""".strip()


    return (
        system_message,
        user_message,
    )


# ---------------------------------------------------------
# Model call.
# ---------------------------------------------------------

def call_semantic_judge(
    system_message,
    user_message,
):
    response = (
        pilot_client
        .chat
        .completions
        .create(
            model=(
                pilot_chat_model
            ),
            messages=[
                {
                    "role": (
                        "system"
                    ),
                    "content": (
                        system_message
                    ),
                },
                {
                    "role": (
                        "user"
                    ),
                    "content": (
                        user_message
                    ),
                },
            ],
            temperature=(
                SEMANTIC_TEMPERATURE
            ),
            max_tokens=(
                SEMANTIC_MAX_TOKENS
            ),
        )
    )


    content = (
        response
        .choices[0]
        .message
        .content
    )


    if not isinstance(
        content,
        str,
    ):
        raise ValueError(
            "Semantic judge returned "
            "non-string content."
        )


    return content


# ---------------------------------------------------------
# Deterministic validation.
# ---------------------------------------------------------

def validate_semantic_payload(
    batch_items,
    payload,
):
    if not isinstance(
        payload,
        dict,
    ):
        raise ValueError(
            "Semantic payload must "
            "be an object."
        )


    judgments = payload.get(
        "judgments"
    )


    if not isinstance(
        judgments,
        list,
    ):
        raise ValueError(
            "'judgments' must be a list."
        )


    expected_ids = {
        item[
            "raw_claim_id"
        ]
        for item in (
            batch_items
        )
    }


    observed_ids = [
        item.get(
            "raw_claim_id"
        )
        if isinstance(
            item,
            dict,
        )
        else None
        for item in (
            judgments
        )
    ]


    if (
        len(
            observed_ids
        )
        != len(
            set(
                observed_ids
            )
        )
    ):
        raise ValueError(
            "Duplicate raw_claim_id "
            "inside semantic response."
        )


    if (
        set(
            observed_ids
        )
        != expected_ids
    ):
        raise ValueError(
            "Semantic response Claim IDs "
            "do not exactly match batch."
        )


    original_by_id = {
        item[
            "raw_claim_id"
        ]: item
        for item in (
            batch_items
        )
    }


    validated = []


    for judgment in judgments:

        if not isinstance(
            judgment,
            dict,
        ):
            raise ValueError(
                "Semantic judgment "
                "must be an object."
            )


        raw_claim_id = judgment[
            "raw_claim_id"
        ]

        decision = str(
            judgment.get(
                "decision",
                "",
            )
        ).strip()


        if (
            decision
            not in SEMANTIC_DECISIONS
        ):
            raise ValueError(
                "Invalid semantic decision: "
                f"{decision!r}"
            )


        checks = judgment.get(
            "checks"
        )


        if not isinstance(
            checks,
            dict,
        ):
            raise ValueError(
                "Semantic checks "
                "must be an object."
            )


        if (
            set(
                checks
            )
            != SEMANTIC_CHECK_NAMES
        ):
            raise ValueError(
                "Semantic check keys "
                "do not match Contract."
            )


        for (
            check_name,
            check_value,
        ) in checks.items():

            if (
                check_value
                not in SEMANTIC_CHECK_VALUES
            ):
                raise ValueError(
                    "Invalid semantic check "
                    f"value for {check_name}: "
                    f"{check_value!r}"
                )


        issues = judgment.get(
            "issues",
            []
        )


        if not isinstance(
            issues,
            list,
        ):
            raise ValueError(
                "Semantic issues must "
                "be a list."
            )


        if (
            len(
                issues
            )
            != len(
                set(
                    issues
                )
            )
        ):
            raise ValueError(
                "Duplicate Semantic Issue "
                "code detected."
            )


        unknown_issues = (
            set(
                issues
            )
            - SEMANTIC_ISSUES
        )


        if unknown_issues:
            raise ValueError(
                "Unknown Semantic Issue codes: "
                f"{sorted(unknown_issues)}"
            )


        repaired_claim = (
            judgment.get(
                "repaired_claim_text_fa"
            )
        )


        recommended_claim_type = (
            judgment.get(
                "recommended_claim_type"
            )
        )


        brief_reason = str(
            judgment.get(
                "brief_reason_fa",
                "",
            )
        ).strip()


        if not brief_reason:
            raise ValueError(
                "Semantic judgment has "
                "an empty brief_reason_fa."
            )


        original = (
            original_by_id[
                raw_claim_id
            ]
        )


        if (
            "SOURCE_METADATA_NOT_DOMAIN_KNOWLEDGE"
            in issues
            and checks[
                "domain_knowledge"
            ] != "FAIL"
        ):
            raise ValueError(
                "SOURCE_METADATA issue requires "
                "domain_knowledge=FAIL."
            )


        if (
            "EVIDENCE_INCOMPLETE"
            in issues
            and checks[
                "evidence_sufficiency"
            ] != "FAIL"
        ):
            raise ValueError(
                "EVIDENCE_INCOMPLETE requires "
                "evidence_sufficiency=FAIL."
            )


        if (
            "UNSUPPORTED_OR_OVERSTATED"
            in issues
            and checks[
                "entailment"
            ] != "FAIL"
        ):
            raise ValueError(
                "UNSUPPORTED issue requires "
                "entailment=FAIL."
            )


        if decision == "ACCEPT":

            if any(
                value != "PASS"
                for value in (
                    checks.values()
                )
            ):
                raise ValueError(
                    "ACCEPT requires "
                    "all checks PASS."
                )


            if repaired_claim is not None:
                raise ValueError(
                    "ACCEPT may not include "
                    "a repaired Claim."
                )


            if (
                recommended_claim_type
                != original[
                    "claim_type"
                ]
            ):
                raise ValueError(
                    "ACCEPT must preserve "
                    "the original Claim Type."
                )


        elif decision == "REPAIR":

            if (
                checks[
                    "entailment"
                ]
                != "PASS"
            ):
                raise ValueError(
                    "REPAIR requires "
                    "entailment=PASS."
                )


            if (
                checks[
                    "evidence_sufficiency"
                ]
                != "PASS"
            ):
                raise ValueError(
                    "REPAIR requires "
                    "evidence_sufficiency=PASS."
                )


            if (
                checks[
                    "domain_knowledge"
                ]
                != "PASS"
            ):
                raise ValueError(
                    "REPAIR requires "
                    "domain_knowledge=PASS."
                )


            if (
                not isinstance(
                    repaired_claim,
                    str,
                )
                or not repaired_claim.strip()
            ):
                raise ValueError(
                    "REPAIR requires "
                    "repaired_claim_text_fa."
                )


            repaired_claim = (
                repaired_claim.strip()
            )


            if (
                recommended_claim_type
                not in ALLOWED_CLAIM_TYPES
            ):
                raise ValueError(
                    "REPAIR recommended Claim Type "
                    "is invalid."
                )


        else:

            if (
                repaired_claim
                is not None
            ):
                raise ValueError(
                    "REJECT must not publish "
                    "a repaired Claim."
                )


            if (
                recommended_claim_type
                is not None
            ):
                raise ValueError(
                    "REJECT must use "
                    "recommended_claim_type=null."
                )


        validated.append({
            "raw_claim_id": (
                raw_claim_id
            ),
            "decision": (
                decision
            ),
            "checks": {
                key: checks[
                    key
                ]
                for key in sorted(
                    SEMANTIC_CHECK_NAMES
                )
            },
            "issues": sorted(
                issues
            ),
            "repaired_claim_text_fa": (
                repaired_claim
            ),
            "recommended_claim_type": (
                recommended_claim_type
            ),
            "brief_reason_fa": (
                brief_reason
            ),
            "original_claim_text_fa": (
                original[
                    "claim_text_fa"
                ]
            ),
            "original_claim_type": (
                original[
                    "claim_type"
                ]
            ),
            "authority_class": (
                original[
                    "authority_class"
                ]
            ),
            "claim_discovery_role": (
                original[
                    "claim_discovery_role"
                ]
            ),
            "semantic_model": (
                pilot_chat_model
            ),
            "semantic_prompt_version": (
                SEMANTIC_PROMPT_VERSION
            ),
        })


    return sorted(
        validated,
        key=lambda item: (
            item[
                "raw_claim_id"
            ]
        ),
    )


# ---------------------------------------------------------
# Batch execution with validated cache + adaptive retry.
# ---------------------------------------------------------

semantic_cache_root.mkdir(
    parents=True,
    exist_ok=True,
)


semantic_judgments = []

semantic_failures = []

semantic_api_calls_this_run = 0
semantic_cache_hits_this_run = 0


print(
    "=== NB12i PILOT SEMANTIC QUALITY GATE ==="
)

print()
print(
    "Pilot Claims:",
    len(
        semantic_items
    ),
)

print(
    "Semantic batches:",
    len(
        semantic_batches
    ),
)

print(
    "Batch size:",
    SEMANTIC_BATCH_SIZE,
)

print()


for batch_number, batch_items in enumerate(
    semantic_batches,
    1,
):

    batch_id = (
        stable_semantic_batch_id(
            batch_items
        )
    )


    cache_path = (
        semantic_cache_root
        / (
            batch_id.replace(
                ":",
                "_",
            )
            + ".json"
        )
    )


    batch_ids = [
        item[
            "raw_claim_id"
        ]
        for item in (
            batch_items
        )
    ]


    print(
        f"[{batch_number:02d}/{len(semantic_batches):02d}]",
        batch_id,
        "| claims:",
        len(
            batch_items
        ),
    )


    if cache_path.exists():

        cached = load_json(
            cache_path
        )


        assert (
            cached.get(
                "batch_id"
            )
            == batch_id
        )


        assert (
            cached.get(
                "semantic_prompt_version"
            )
            == SEMANTIC_PROMPT_VERSION
        )


        assert (
            cached.get(
                "model"
            )
            == pilot_chat_model
        )


        cached_judgments = (
            cached.get(
                "judgments"
            )
        )


        assert isinstance(
            cached_judgments,
            list,
        )


        assert (
            {
                item[
                    "raw_claim_id"
                ]
                for item in (
                    cached_judgments
                )
            }
            == set(
                batch_ids
            )
        )


        semantic_judgments.extend(
            cached_judgments
        )


        semantic_cache_hits_this_run += 1


        print(
            "   cache: HIT"
        )

        continue


    validated_judgments = None
    last_error = None


    for attempt in range(
        1,
        SEMANTIC_MAX_ATTEMPTS + 1,
    ):

        (
            system_message,
            user_message,
        ) = build_semantic_prompt(
            batch_items,
            retry_feedback=(
                last_error
            ),
        )


        try:

            content = (
                call_semantic_judge(
                    system_message,
                    user_message,
                )
            )


            semantic_api_calls_this_run += 1


            payload = (
                parse_model_json(
                    content
                )
            )


            validated_judgments = (
                validate_semantic_payload(
                    batch_items,
                    payload,
                )
            )


            break


        except Exception as exc:

            last_error = (
                f"{type(exc).__name__}: "
                f"{exc}"
            )


            print(
                f"   attempt {attempt}:",
                "INVALID",
                "|",
                last_error,
            )


            if (
                attempt
                < SEMANTIC_MAX_ATTEMPTS
            ):
                time.sleep(
                    min(
                        2.0,
                        float(
                            attempt
                        ),
                    )
                )


    if (
        validated_judgments
        is None
    ):

        semantic_failures.append({
            "batch_id": (
                batch_id
            ),
            "raw_claim_ids": (
                batch_ids
            ),
            "last_error": (
                last_error
            ),
        })


        print(
            "   result: FAIL"
        )

        continue


    cache_payload = {
        "batch_id": (
            batch_id
        ),
        "semantic_prompt_version": (
            SEMANTIC_PROMPT_VERSION
        ),
        "model": (
            pilot_chat_model
        ),
        "raw_claim_ids": (
            batch_ids
        ),
        "judgments": (
            validated_judgments
        ),
    }


    cache_text = (
        json.dumps(
            cache_payload,
            ensure_ascii=False,
            indent=2,
            sort_keys=True,
        )
        + "\n"
    )


    with cache_path.open(
        "x",
        encoding="utf-8",
    ) as file:
        file.write(
            cache_text
        )


    semantic_judgments.extend(
        validated_judgments
    )


    print(
        "   result: VALID"
    )


print()


assert not semantic_failures, (
    "Semantic Adjudication has unresolved "
    "batch failures:\n"
    + json.dumps(
        semantic_failures,
        ensure_ascii=False,
        indent=2,
    )
)


semantic_judgments = sorted(
    semantic_judgments,
    key=lambda item: (
        item[
            "raw_claim_id"
        ]
    ),
)


judged_claim_ids = [
    item[
        "raw_claim_id"
    ]
    for item in (
        semantic_judgments
    )
]


assert (
    len(
        judged_claim_ids
    )
    == 59
)


assert (
    len(
        judged_claim_ids
    )
    == len(
        set(
            judged_claim_ids
        )
    )
)


assert (
    set(
        judged_claim_ids
    )
    == set(
        semantic_claim_by_id
    )
)


# ---------------------------------------------------------
# Aggregate semantic results.
# ---------------------------------------------------------

decision_counts = Counter(
    item[
        "decision"
    ]
    for item in (
        semantic_judgments
    )
)


issue_counts = Counter()

check_fail_counts = Counter()


for judgment in (
    semantic_judgments
):

    for issue in (
        judgment[
            "issues"
        ]
    ):
        issue_counts[
            issue
        ] += 1


    for (
        check_name,
        check_value,
    ) in (
        judgment[
            "checks"
        ].items()
    ):

        if (
            check_value
            == "FAIL"
        ):
            check_fail_counts[
                check_name
            ] += 1


all_window_by_id = {
    window[
        "prompt_window_id"
    ]: window
    for window in (
        prompt_windows
    )
}


decision_by_source = defaultdict(
    Counter
)

decision_by_role = defaultdict(
    Counter
)


for judgment in (
    semantic_judgments
):

    claim = (
        semantic_claim_by_id[
            judgment[
                "raw_claim_id"
            ]
        ]
    )

    window = (
        all_window_by_id[
            claim[
                "prompt_window_id"
            ]
        ]
    )


    decision_by_source[
        window[
            "source_type"
        ]
    ][
        judgment[
            "decision"
        ]
    ] += 1


    decision_by_role[
        claim[
            "claim_discovery_role"
        ]
    ][
        judgment[
            "decision"
        ]
    ] += 1


# ---------------------------------------------------------
# Publish semantic adjudication.
# ---------------------------------------------------------

semantic_adjudication_text = "".join(
    json.dumps(
        item,
        ensure_ascii=False,
        sort_keys=True,
    )
    + "\n"
    for item in (
        semantic_judgments
    )
)


semantic_adjudication_sha256 = (
    hashlib.sha256(
        semantic_adjudication_text.encode(
            "utf-8"
        )
    ).hexdigest()
)


semantic_report = {
    "report_version": (
        "claim_extraction_pilot_semantic_report_v1"
    ),

    "source_pilot_report": (
        pilot_report_path.name
    ),

    "source_raw_claims": (
        pilot_raw_path.name
    ),

    "source_raw_claims_sha256": (
        file_sha256(
            pilot_raw_path
        )
    ),

    "semantic_prompt_version": (
        SEMANTIC_PROMPT_VERSION
    ),

    "semantic_model": (
        pilot_chat_model
    ),

    "claims_adjudicated": (
        len(
            semantic_judgments
        )
    ),

    "decision_counts": dict(
        sorted(
            decision_counts.items()
        )
    ),

    "failed_checks": dict(
        sorted(
            check_fail_counts.items()
        )
    ),

    "issue_counts": dict(
        sorted(
            issue_counts.items()
        )
    ),

    "decisions_by_source_type": {
        source_type: dict(
            sorted(
                counts.items()
            )
        )
        for (
            source_type,
            counts,
        ) in sorted(
            decision_by_source.items()
        )
    },

    "decisions_by_claim_discovery_role": {
        role: dict(
            sorted(
                counts.items()
            )
        )
        for (
            role,
            counts,
        ) in sorted(
            decision_by_role.items()
        )
    },

    "unresolved_batch_failures": 0,

    "semantic_execution_status": (
        "PASS"
    ),

    "full_registry_readiness": (
        "AWAITING_QUALITY_DECISION"
    ),

    "full_registry_extraction_started": (
        False
    ),

    "artifact": {
        "adjudication": (
            semantic_adjudication_path.name
        ),
        "adjudication_sha256": (
            semantic_adjudication_sha256
        ),
    },
}


semantic_report_text = (
    json.dumps(
        semantic_report,
        ensure_ascii=False,
        indent=2,
        sort_keys=True,
    )
    + "\n"
)


def publish_semantic_no_overwrite(
    path,
    content,
):
    if path.exists():

        existing_text = (
            path.read_text(
                encoding="utf-8"
            )
        )


        assert (
            existing_text
            == content
        ), (
            "Existing Semantic Artifact differs "
            f"from staged output: {path.name}. "
            "Refusing overwrite."
        )


        return (
            "REUSED_EXISTING_IDENTICAL"
        )


    with path.open(
        "x",
        encoding="utf-8",
    ) as file:
        file.write(
            content
        )


    return (
        "CREATED_NEW"
    )


semantic_adjudication_write_status = (
    publish_semantic_no_overwrite(
        semantic_adjudication_path,
        semantic_adjudication_text,
    )
)


semantic_report_write_status = (
    publish_semantic_no_overwrite(
        semantic_report_path,
        semantic_report_text,
    )
)


disk_semantic_judgments = load_jsonl(
    semantic_adjudication_path
)

disk_semantic_report = load_json(
    semantic_report_path
)


assert (
    disk_semantic_judgments
    == semantic_judgments
)


assert (
    disk_semantic_report
    == semantic_report
)


assert (
    file_sha256(
        semantic_adjudication_path
    )
    == semantic_adjudication_sha256
)


pilot_semantic_adjudication_status = (
    "PASS"
)


# ---------------------------------------------------------
# Human-readable result.
# ---------------------------------------------------------

print(
    "=== NB12i PILOT SEMANTIC QUALITY RESULT ==="
)

print()
print(
    "Claims adjudicated:",
    len(
        semantic_judgments
    ),
)

print(
    "Semantic API calls this run:",
    semantic_api_calls_this_run,
)

print(
    "Validated semantic cache hits:",
    semantic_cache_hits_this_run,
)


print()
print(
    "Decisions:",
    dict(
        sorted(
            decision_counts.items()
        )
    ),
)


print()
print(
    "Failed checks:",
    dict(
        sorted(
            check_fail_counts.items()
        )
    ),
)


print()
print(
    "Issue counts:",
    dict(
        sorted(
            issue_counts.items()
        )
    ),
)


print()
print(
    "Decisions by source type:"
)

for (
    source_type,
    counts,
) in sorted(
    decision_by_source.items()
):

    print(
        "  -",
        source_type,
        dict(
            sorted(
                counts.items()
            )
        ),
    )


print()
print(
    "Decisions by authority role:"
)

for (
    role,
    counts,
) in sorted(
    decision_by_role.items()
):

    print(
        "  -",
        role,
        dict(
            sorted(
                counts.items()
            )
        ),
    )


print()
print(
    "=== REPAIR CLAIMS ==="
)


repair_items = [
    item
    for item in (
        semantic_judgments
    )
    if item[
        "decision"
    ] == "REPAIR"
]


if not repair_items:
    print(
        "  None"
    )


for item in (
    repair_items[:20]
):

    print()
    print(
        "Raw Claim ID:",
        item[
            "raw_claim_id"
        ],
    )

    print(
        "Original:",
        item[
            "original_claim_text_fa"
        ],
    )

    print(
        "Repair:",
        item[
            "repaired_claim_text_fa"
        ],
    )

    print(
        "Type:",
        item[
            "original_claim_type"
        ],
        "->",
        item[
            "recommended_claim_type"
        ],
    )

    print(
        "Issues:",
        item[
            "issues"
        ],
    )

    print(
        "Reason:",
        item[
            "brief_reason_fa"
        ],
    )


if len(
    repair_items
) > 20:
    print(
        "  ...",
        len(
            repair_items
        ) - 20,
        "additional REPAIR judgments not printed",
    )


print()
print(
    "=== REJECT CLAIMS ==="
)


reject_items = [
    item
    for item in (
        semantic_judgments
    )
    if item[
        "decision"
    ] == "REJECT"
]


if not reject_items:
    print(
        "  None"
    )


for item in (
    reject_items[:20]
):

    print()
    print(
        "Raw Claim ID:",
        item[
            "raw_claim_id"
        ],
    )

    print(
        "Claim:",
        item[
            "original_claim_text_fa"
        ],
    )

    print(
        "Issues:",
        item[
            "issues"
        ],
    )

    print(
        "Failed checks:",
        [
            key
            for (
                key,
                value,
            ) in (
                item[
                    "checks"
                ].items()
            )
            if value == "FAIL"
        ],
    )

    print(
        "Reason:",
        item[
            "brief_reason_fa"
        ],
    )


if len(
    reject_items
) > 20:
    print(
        "  ...",
        len(
            reject_items
        ) - 20,
        "additional REJECT judgments not printed",
    )


print()
print(
    "=== ACCEPT SAMPLE ==="
)


accept_items = [
    item
    for item in (
        semantic_judgments
    )
    if item[
        "decision"
    ] == "ACCEPT"
]


for item in (
    accept_items[:8]
):

    print()
    print(
        "Raw Claim ID:",
        item[
            "raw_claim_id"
        ],
    )

    print(
        "Claim:",
        item[
            "original_claim_text_fa"
        ],
    )

    print(
        "Reason:",
        item[
            "brief_reason_fa"
        ],
    )


print()
print(
    "Semantic Adjudication Artifact:",
    semantic_adjudication_path.relative_to(
        project_root
    ),
)

print(
    "Semantic Adjudication SHA-256:",
    semantic_adjudication_sha256,
)

print(
    "Write status:",
    semantic_adjudication_write_status,
)


print()
print(
    "Semantic Report:",
    semantic_report_path.relative_to(
        project_root
    ),
)

print(
    "Report write status:",
    semantic_report_write_status,
)


print()
print(
    "Full Registry readiness:",
    "AWAITING_QUALITY_DECISION",
)

print(
    "Full Registry Extraction:",
    "NOT_STARTED",
)


print()
print(
    "CODE CELL 11 GATE:",
    pilot_semantic_adjudication_status,
)
```

<ul>
<li>برخی Fragmentها یا جمله‌های وابسته به Context به‌عنوان Claim تولید شده‌اند؛</li>
<li>بعضی Claimها فاعل، موضوع یا شیء مشخص ندارند؛</li>

<li>در بعضی Claimهای عددی، عدد بدون موضوع کامل استخراج شده است؛</li>
<li>در چند مورد Claim Type با ماهیت گزاره سازگار نبوده است؛</li>

<h3>هدف Prompt v2</h3>

<p>
Prompt جدید باید قبل از تولید Claim، Eligibility معنایی گزاره را سخت‌گیرانه‌تر بررسی کند.
هدف این نیست که تعداد Claimها حتماً زیاد بماند؛ هدف این است که Claimهای کم‌ارزش یا ناقص اصلاً وارد Raw Registry نشوند.
</p>

<h3>قواعد جدید Extraction</h3>

<ul>
<li>Claim باید بدون Context بیرونی قابل فهم باشد؛</li>
<li>فاعل، موضوع و رابطهٔ اصلی Claim باید مشخص باشند؛</li>
<li>عباراتی مانند «این»، «آن»، «او»، «می‌گوید»، «بهش» و Fragmentهای مشابه فقط وقتی مجازند که مرجع آن‌ها داخل خود Claim روشن شده باشد؛</li>
<li>List Header به‌تنهایی Claim نیست؛ اگر گفته می‌شود «دو ابزار وجود دارد»، خود ابزارها نیز باید در Claim و Evidence موجود باشند؛</li>
<li>Source Metadata، Editorial Commentary، معرفی کتاب، توضیح پرسشنامه و اطلاعات مشابه دانش دامنه محسوب نمی‌شوند؛</li>
<li>نقل‌قول ادبی یا اخلاقی عمومی فقط وقتی Claim مالی محسوب می‌شود که دانش مالی صریح و مستقل ارائه کند؛</li>
<li>Claim عددی باید موضوع، مقدار و معنای عدد را مشخص کند؛</li>
<li>هر Fact موجود در Claim باید توسط Evidenceهای انتخاب‌شده پوشش داده شود؛</li>
<li>اگر فاعل از یک Slice و گزاره از Slice دیگر آمده است، هر دو Slice باید به‌عنوان Evidence معتبر ذکر شوند؛</li>
<li>اگر Evidence کامل وجود ندارد، Claim نباید تولید شود؛</li>
<li>در صورت تردید، عدم تولید Claim بر تولید Claim ناقص ترجیح دارد.</li>
</ul>

<h3>تعریف دقیق‌تر Claim Typeها</h3>

<ul>
<li><code>DEFINITION</code> — تعریف معنی یک مفهوم؛</li>
<li><code>RELATION</code> — ارتباط غیرعلّی میان دو یا چند مفهوم؛</li>
<li><code>CAUSAL</code> — رابطهٔ علت و معلولی صریح؛</li>
<li><code>RULE_OR_PRINCIPLE</code> — قاعده، اصل یا توصیهٔ کلی؛</li>
<li><code>PROCEDURE_OR_ACTION</code> — اقدام یا مرحلهٔ عملی مشخص؛</li>
<li><code>COMPARISON</code> — مقایسهٔ صریح میان گزینه‌ها یا مفاهیم؛</li>
<li><code>CONDITION_OR_CONSTRAINT</code> — شرط، محدودیت یا الزام؛</li>
<li><code>CLASSIFICATION</code> — دسته‌بندی یا عضویت در یک گروه؛</li>
<li><code>NUMERIC_OR_QUANTITATIVE</code> — گزاره‌ای که مقدار عددی بخش اصلی معنای آن است؛</li>
<li><code>OTHER_ATOMIC</code> — فقط برای گزارهٔ مستقل و معتبر که در دسته‌های دیگر قرار نمی‌گیرد.</li>
</ul>

```python
# Code Cell 12

assert (
    pilot_semantic_adjudication_status
    == "PASS"
), (
    "Code Cell 11 must PASS before "
    "repairing the extraction prompt."
)


PILOT_V2_VERSION = (
    "claim_extraction_pilot_v2"
)

PILOT_V2_PROMPT_VERSION = (
    "claim_extraction_prompt_v2"
)

PILOT_V2_MAX_ATTEMPTS = 3
PILOT_V2_MAX_TOKENS = 2500
PILOT_V2_TEMPERATURE = 0


pilot_v2_raw_path = (
    registry_artifact_root
    / "claim_extraction_pilot_raw_v2.jsonl"
)

pilot_v2_evidence_path = (
    registry_artifact_root
    / "claim_extraction_pilot_evidence_v2.jsonl"
)

pilot_v2_report_path = (
    registry_artifact_root
    / "claim_extraction_pilot_report_v2.json"
)

pilot_v2_cache_root = (
    registry_artifact_root
    / "claim_extraction_pilot_cache_v2"
)


# ---------------------------------------------------------
# Ensure the controlled comparison uses exactly
# the same 15 Prompt Windows as Pilot v1.
# ---------------------------------------------------------

pilot_v1_report_from_disk = load_json(
    pilot_report_path
)


pilot_v1_window_ids = (
    pilot_v1_report_from_disk[
        "selection_contract"
    ][
        "window_ids"
    ]
)


pilot_v2_window_ids = [
    window[
        "prompt_window_id"
    ]
    for window in (
        pilot_windows
    )
]


assert (
    pilot_v2_window_ids
    == pilot_v1_window_ids
), (
    "Pilot v2 must use exactly the same "
    "Prompt Windows as Pilot v1."
)


# ---------------------------------------------------------
# Stricter extraction prompt.
# ---------------------------------------------------------

def build_claim_prompt_v2(
    window,
    context_text,
    slice_map,
    retry_feedback=None,
):
    eligible_slice_ids = [
        slice_id
        for (
            slice_id,
            item,
        ) in (
            slice_map.items()
        )
        if item[
            "evidence_allowed"
        ]
    ]


    claim_type_definitions = {
        "DEFINITION": (
            "Defines what a concept means."
        ),
        "RELATION": (
            "States a non-causal relationship "
            "between identifiable concepts."
        ),
        "CAUSAL": (
            "States an explicit cause-effect "
            "relationship."
        ),
        "RULE_OR_PRINCIPLE": (
            "States a general rule, principle, "
            "or explicit general recommendation."
        ),
        "PROCEDURE_OR_ACTION": (
            "States a concrete action or "
            "procedural step."
        ),
        "COMPARISON": (
            "Explicitly compares identifiable "
            "options or concepts."
        ),
        "CONDITION_OR_CONSTRAINT": (
            "States a condition, requirement, "
            "restriction, or constraint."
        ),
        "CLASSIFICATION": (
            "Classifies an identifiable entity "
            "into a category or set."
        ),
        "NUMERIC_OR_QUANTITATIVE": (
            "A numeric value or quantity is "
            "essential to the Claim meaning."
        ),
        "OTHER_ATOMIC": (
            "A complete standalone atomic "
            "domain Claim that does not fit "
            "the other categories."
        ),
    }


    retry_block = ""


    if retry_feedback:

        retry_block = f"""

IMPORTANT RETRY CORRECTION

The previous response failed deterministic validation.

Validator error:
{retry_feedback}

Return the COMPLETE JSON object again.

Correct the exact validation problem.
Do not loosen Evidence requirements.
Do not invent new facts.
"""


    system_message = """
You are performing high-precision extraction for a Persian financial knowledge registry.

Your task is NOT to maximize the number of claims.
Your task is to extract only high-value, standalone, evidence-complete domain knowledge.

Use ONLY the supplied source slices.
Never use outside knowledge.

A valid Claim must satisfy ALL of these conditions:

1. It is a complete standalone knowledge statement.
2. A reader can understand it without unseen conversation context.
3. Its subject/entity is explicit.
4. Its main predicate/relation is explicit.
5. It is atomic enough to represent one knowledge unit.
6. Every factual component is directly supported by the cited Evidence.
7. It is genuine domain knowledge, not source/editorial metadata.
8. It is not merely a fragment, instruction fragment, heading, list introduction, quotation attribution, or vague conversational phrase.
9. Its claim_type accurately matches its semantics.

Do NOT extract:
- "this/that/he/she says..." claims with unresolved references,
- source commentary such as descriptions of a book, questionnaire, edition, author style, or editorial adaptation,
- list-introduction claims such as "there are two tools" unless the tools themselves are explicitly identified and supported,
- numeric fragments where the quantity's subject is not explicit,
- incomplete imperatives,
- vague conversational fragments,
- general literary/moral sayings unless they explicitly state useful financial/economic knowledge,
- claims whose complete meaning requires evidence that cannot be cited.

If the subject comes from one eligible slice and the predicate from another eligible slice, cite BOTH.
If a required supporting slice is not evidence-eligible, do not extract that Claim.

Prefer returning fewer high-quality Claims over uncertain Claims.

Return JSON only.
No markdown.
No explanation outside JSON.
""".strip()


    user_message = f"""
Extract at most {MAX_CLAIMS_PER_WINDOW} high-quality atomic Claims.

Allowed claim types and meanings:

{json.dumps(
    claim_type_definitions,
    ensure_ascii=False,
    indent=2,
)}

Eligible evidence slice_ids:

{json.dumps(
    eligible_slice_ids,
    ensure_ascii=False,
)}

Required JSON schema:

{{
  "claims": [
    {{
      "claim_text_fa": "یک گزاره مستقل، کامل و دقیق به فارسی",
      "claim_type": "ONE_ALLOWED_TYPE",
      "evidence": [
        {{
          "slice_id": "S01",
          "quote_exact": "متن دقیق و بدون بازنویسی از همان Slice"
        }}
      ]
    }}
  ]
}}

Final eligibility checklist BEFORE emitting each Claim:

- Is the subject explicit?
- Is the Claim understandable outside this source?
- Is it useful financial/economic/personal-finance domain knowledge?
- Is it more than source metadata or editorial commentary?
- Is it more than a fragment or list heading?
- Does the Evidence support EVERY meaningful part of the Claim?
- If the Claim contains a number, is the number's subject explicit?
- If the Claim contains a causal relation, are both cause and effect explicit in Evidence?
- If the Claim contains a classification, is the classified entity explicit?
- Is the selected claim_type semantically correct?

If ANY answer is no, do not emit that Claim.

Evidence rules:

1. Evidence may use only eligible slice_ids.
2. Never use evidence_allowed=false slices.
3. quote_exact must be copied verbatim.
4. Do not insert ellipses.
5. Do not paraphrase Evidence.
6. Prefer a short distinctive quote, but it must support the entire Claim.
7. Use multiple Evidence items when needed for the complete meaning.
8. If complete Evidence is unavailable, omit the Claim.

Window metadata:

source_type={window['source_type']}
authority_class={window['authority_class']}
claim_discovery_role={window['claim_discovery_role']}

SOURCE SLICES:

{context_text}

{retry_block}
""".strip()


    return (
        system_message,
        user_message,
    )


# ---------------------------------------------------------
# v2 model call.
# ---------------------------------------------------------

def call_pilot_model_v2(
    system_message,
    user_message,
):
    response = (
        pilot_client
        .chat
        .completions
        .create(
            model=(
                pilot_chat_model
            ),
            messages=[
                {
                    "role": "system",
                    "content": (
                        system_message
                    ),
                },
                {
                    "role": "user",
                    "content": (
                        user_message
                    ),
                },
            ],
            temperature=(
                PILOT_V2_TEMPERATURE
            ),
            max_tokens=(
                PILOT_V2_MAX_TOKENS
            ),
        )
    )


    content = (
        response
        .choices[0]
        .message
        .content
    )


    if not isinstance(
        content,
        str,
    ):
        raise ValueError(
            "Pilot v2 model response "
            "content is not a string."
        )


    return content


# ---------------------------------------------------------
# Reuse the proven deterministic Evidence validator,
# then explicitly mark outputs as Prompt v2.
# ---------------------------------------------------------

def validate_and_build_pilot_bundle_v2(
    window,
    payload,
    slice_map,
):
    bundle = (
        validate_and_build_pilot_bundle(
            window,
            payload,
            slice_map,
        )
    )


    for claim in (
        bundle[
            "raw_claims"
        ]
    ):
        claim[
            "extraction_version"
        ] = PILOT_V2_VERSION

        claim[
            "extraction_prompt_version"
        ] = (
            PILOT_V2_PROMPT_VERSION
        )


    for evidence in (
        bundle[
            "evidence_links"
        ]
    ):
        evidence[
            "extraction_version"
        ] = PILOT_V2_VERSION

        evidence[
            "extraction_prompt_version"
        ] = (
            PILOT_V2_PROMPT_VERSION
        )


    return bundle


# ---------------------------------------------------------
# Run same 15 Window controlled Pilot.
# ---------------------------------------------------------

pilot_v2_cache_root.mkdir(
    parents=True,
    exist_ok=True,
)


pilot_v2_bundles = []

pilot_v2_failures = []

pilot_v2_api_calls_this_run = 0
pilot_v2_cache_hits_this_run = 0


print(
    "=== NB12i CLAIM EXTRACTION PILOT V2 ==="
)

print()
print(
    "Prompt version:",
    PILOT_V2_PROMPT_VERSION,
)

print(
    "Model:",
    pilot_chat_model,
)

print(
    "Controlled Prompt Windows:",
    len(
        pilot_windows
    ),
)

print()


for index, window in enumerate(
    pilot_windows,
    1,
):

    cache_path = (
        pilot_v2_cache_root
        / (
            window[
                "prompt_window_id"
            ].replace(
                ":",
                "_",
            )
            + ".json"
        )
    )


    print(
        f"[{index:02d}/{len(pilot_windows):02d}]",
        window[
            "source_type"
        ],
        "| role:",
        window[
            "claim_discovery_role"
        ],
        "| chars:",
        window[
            "raw_text_chars"
        ],
        "|",
        window[
            "prompt_window_id"
        ],
    )


    if cache_path.exists():

        cached = load_json(
            cache_path
        )


        assert (
            cached.get(
                "prompt_version"
            )
            == PILOT_V2_PROMPT_VERSION
        )


        assert (
            cached.get(
                "model"
            )
            == pilot_chat_model
        )


        assert (
            cached.get(
                "prompt_window_id"
            )
            == window[
                "prompt_window_id"
            ]
        )


        cached_bundle = (
            cached[
                "bundle"
            ]
        )


        pilot_v2_bundles.append(
            cached_bundle
        )


        pilot_v2_cache_hits_this_run += 1


        print(
            "   cache: HIT",
            "| claims:",
            len(
                cached_bundle[
                    "raw_claims"
                ]
            ),
        )


        continue


    (
        context_text,
        slice_map,
    ) = materialize_pilot_window(
        window
    )


    validated_bundle = None
    last_error = None


    for attempt in range(
        1,
        PILOT_V2_MAX_ATTEMPTS + 1,
    ):

        (
            system_message,
            user_message,
        ) = build_claim_prompt_v2(
            window,
            context_text,
            slice_map,
            retry_feedback=(
                last_error
            ),
        )


        try:

            content = (
                call_pilot_model_v2(
                    system_message,
                    user_message,
                )
            )


            pilot_v2_api_calls_this_run += 1


            payload = (
                parse_model_json(
                    content
                )
            )


            validated_bundle = (
                validate_and_build_pilot_bundle_v2(
                    window,
                    payload,
                    slice_map,
                )
            )


            break


        except Exception as exc:

            last_error = (
                f"{type(exc).__name__}: "
                f"{exc}"
            )


            print(
                f"   attempt {attempt}:",
                "INVALID",
                "|",
                last_error,
            )


            if (
                attempt
                < PILOT_V2_MAX_ATTEMPTS
            ):
                time.sleep(
                    min(
                        2.0,
                        float(
                            attempt
                        ),
                    )
                )


    if (
        validated_bundle
        is None
    ):

        pilot_v2_failures.append({
            "prompt_window_id": (
                window[
                    "prompt_window_id"
                ]
            ),
            "source_type": (
                window[
                    "source_type"
                ]
            ),
            "last_error": (
                last_error
            ),
        })


        print(
            "   result: FAIL"
        )


        continue


    cache_payload = {
        "prompt_version": (
            PILOT_V2_PROMPT_VERSION
        ),
        "model": (
            pilot_chat_model
        ),
        "prompt_window_id": (
            window[
                "prompt_window_id"
            ]
        ),
        "bundle": (
            validated_bundle
        ),
    }


    cache_text = (
        json.dumps(
            cache_payload,
            ensure_ascii=False,
            indent=2,
            sort_keys=True,
        )
        + "\n"
    )


    with cache_path.open(
        "x",
        encoding="utf-8",
    ) as file:
        file.write(
            cache_text
        )


    pilot_v2_bundles.append(
        validated_bundle
    )


    print(
        "   result: VALID",
        "| claims:",
        len(
            validated_bundle[
                "raw_claims"
            ]
        ),
    )


print()


assert not pilot_v2_failures, (
    "Pilot v2 has unresolved Window failures:\n"
    + json.dumps(
        pilot_v2_failures,
        ensure_ascii=False,
        indent=2,
    )
)


assert (
    len(
        pilot_v2_bundles
    )
    == len(
        pilot_windows
    )
)


# ---------------------------------------------------------
# Aggregate.
# ---------------------------------------------------------

pilot_v2_raw_claims = sorted(
    [
        claim
        for bundle in (
            pilot_v2_bundles
        )
        for claim in (
            bundle[
                "raw_claims"
            ]
        )
    ],
    key=lambda item: (
        item[
            "raw_claim_id"
        ]
    ),
)


pilot_v2_evidence_links = sorted(
    [
        evidence
        for bundle in (
            pilot_v2_bundles
        )
        for evidence in (
            bundle[
                "evidence_links"
            ]
        )
    ],
    key=lambda item: (
        item[
            "evidence_id"
        ]
    ),
)


pilot_v2_raw_claim_ids = [
    item[
        "raw_claim_id"
    ]
    for item in (
        pilot_v2_raw_claims
    )
]


pilot_v2_evidence_ids = [
    item[
        "evidence_id"
    ]
    for item in (
        pilot_v2_evidence_links
    )
]


assert (
    len(
        pilot_v2_raw_claim_ids
    )
    == len(
        set(
            pilot_v2_raw_claim_ids
        )
    )
), (
    "Duplicate Pilot v2 raw_claim_id."
)


assert (
    len(
        pilot_v2_evidence_ids
    )
    == len(
        set(
            pilot_v2_evidence_ids
        )
    )
), (
    "Duplicate Pilot v2 evidence_id."
)


pilot_v2_window_by_id = {
    window[
        "prompt_window_id"
    ]: window
    for window in (
        pilot_windows
    )
}


pilot_v2_claims_by_source = Counter()

pilot_v2_claims_by_role = Counter()

pilot_v2_claims_by_type = Counter()

pilot_v2_claims_by_window = Counter()


for claim in (
    pilot_v2_raw_claims
):

    window = (
        pilot_v2_window_by_id[
            claim[
                "prompt_window_id"
            ]
        ]
    )


    pilot_v2_claims_by_source[
        window[
            "source_type"
        ]
    ] += 1


    pilot_v2_claims_by_role[
        claim[
            "claim_discovery_role"
        ]
    ] += 1


    pilot_v2_claims_by_type[
        claim[
            "claim_type"
        ]
    ] += 1


    pilot_v2_claims_by_window[
        claim[
            "prompt_window_id"
        ]
    ] += 1


for source_type in [
    "book",
    "transcript",
    "website",
]:

    assert (
        pilot_v2_claims_by_source[
            source_type
        ]
        > 0
    ), (
        "Pilot v2 produced no Claims "
        f"for {source_type!r}."
    )


pilot_v2_zero_claim_windows = [
    window[
        "prompt_window_id"
    ]
    for window in (
        pilot_windows
    )
    if (
        pilot_v2_claims_by_window[
            window[
                "prompt_window_id"
            ]
        ]
        == 0
    )
]


# ---------------------------------------------------------
# Publish v2 Pilot artifacts.
# ---------------------------------------------------------

pilot_v2_raw_text = "".join(
    json.dumps(
        item,
        ensure_ascii=False,
        sort_keys=True,
    )
    + "\n"
    for item in (
        pilot_v2_raw_claims
    )
)


pilot_v2_evidence_text = "".join(
    json.dumps(
        item,
        ensure_ascii=False,
        sort_keys=True,
    )
    + "\n"
    for item in (
        pilot_v2_evidence_links
    )
)


pilot_v2_raw_sha256 = (
    hashlib.sha256(
        pilot_v2_raw_text.encode(
            "utf-8"
        )
    ).hexdigest()
)


pilot_v2_evidence_sha256 = (
    hashlib.sha256(
        pilot_v2_evidence_text.encode(
            "utf-8"
        )
    ).hexdigest()
)


baseline_semantic_report = load_json(
    semantic_report_path
)


pilot_v2_report = {
    "report_version": (
        "claim_extraction_pilot_report_v2"
    ),

    "pilot_version": (
        PILOT_V2_VERSION
    ),

    "prompt_version": (
        PILOT_V2_PROMPT_VERSION
    ),

    "model": (
        pilot_chat_model
    ),

    "controlled_comparison": {
        "same_windows_as_pilot_v1": True,
        "window_count": (
            len(
                pilot_windows
            )
        ),
        "pilot_v1_raw_claim_count": (
            len(
                pilot_raw_claims
            )
        ),
        "pilot_v1_semantic_decisions": (
            baseline_semantic_report[
                "decision_counts"
            ]
        ),
    },

    "generation_contract": {
        "temperature": (
            PILOT_V2_TEMPERATURE
        ),
        "max_tokens": (
            PILOT_V2_MAX_TOKENS
        ),
        "max_attempts": (
            PILOT_V2_MAX_ATTEMPTS
        ),
        "adaptive_retry": True,
        "exact_evidence_validation": True,
    },

    "results": {
        "raw_claims": (
            len(
                pilot_v2_raw_claims
            )
        ),
        "evidence_links": (
            len(
                pilot_v2_evidence_links
            )
        ),
        "zero_claim_windows": (
            len(
                pilot_v2_zero_claim_windows
            )
        ),
        "claims_by_source_type": dict(
            sorted(
                pilot_v2_claims_by_source.items()
            )
        ),
        "claims_by_role": dict(
            sorted(
                pilot_v2_claims_by_role.items()
            )
        ),
        "claims_by_type": dict(
            sorted(
                pilot_v2_claims_by_type.items()
            )
        ),
        "unresolved_window_failures": 0,
        "duplicate_raw_claim_ids": 0,
        "duplicate_evidence_ids": 0,
        "structural_grounding_failures": 0,
    },

    "semantic_status": (
        "NOT_YET_RUN"
    ),

    "full_registry_status": (
        "NOT_STARTED"
    ),

    "artifacts": {
        "raw_claims": (
            pilot_v2_raw_path.name
        ),
        "raw_claims_sha256": (
            pilot_v2_raw_sha256
        ),
        "evidence_links": (
            pilot_v2_evidence_path.name
        ),
        "evidence_links_sha256": (
            pilot_v2_evidence_sha256
        ),
    },
}


pilot_v2_report_text = (
    json.dumps(
        pilot_v2_report,
        ensure_ascii=False,
        indent=2,
        sort_keys=True,
    )
    + "\n"
)


pilot_v2_raw_write_status = (
    publish_no_overwrite(
        pilot_v2_raw_path,
        pilot_v2_raw_text,
    )
)


pilot_v2_evidence_write_status = (
    publish_no_overwrite(
        pilot_v2_evidence_path,
        pilot_v2_evidence_text,
    )
)


pilot_v2_report_write_status = (
    publish_no_overwrite(
        pilot_v2_report_path,
        pilot_v2_report_text,
    )
)


# ---------------------------------------------------------
# Independent disk validation.
# ---------------------------------------------------------

disk_pilot_v2_raw = load_jsonl(
    pilot_v2_raw_path
)

disk_pilot_v2_evidence = load_jsonl(
    pilot_v2_evidence_path
)

disk_pilot_v2_report = load_json(
    pilot_v2_report_path
)


assert (
    disk_pilot_v2_raw
    == pilot_v2_raw_claims
)


assert (
    disk_pilot_v2_evidence
    == pilot_v2_evidence_links
)


assert (
    disk_pilot_v2_report
    == pilot_v2_report
)


assert (
    file_sha256(
        pilot_v2_raw_path
    )
    == pilot_v2_raw_sha256
)


assert (
    file_sha256(
        pilot_v2_evidence_path
    )
    == pilot_v2_evidence_sha256
)


pilot_v2_extraction_status = (
    "PASS"
)


# ---------------------------------------------------------
# Human-readable comparison.
# ---------------------------------------------------------

print(
    "=== NB12i CLAIM EXTRACTION PILOT V2 RESULT ==="
)

print()
print(
    "Controlled Prompt Windows:",
    len(
        pilot_windows
    ),
)

print(
    "API calls this run:",
    pilot_v2_api_calls_this_run,
)

print(
    "Validated cache hits:",
    pilot_v2_cache_hits_this_run,
)


print()
print(
    "Pilot v1 Raw Claims:",
    len(
        pilot_raw_claims
    ),
)

print(
    "Pilot v2 Raw Claims:",
    len(
        pilot_v2_raw_claims
    ),
)

print(
    "Pilot v2 Evidence Links:",
    len(
        pilot_v2_evidence_links
    ),
)

print(
    "Pilot v2 zero-Claim Windows:",
    len(
        pilot_v2_zero_claim_windows
    ),
)


print()
print(
    "Claims by source type:",
    dict(
        sorted(
            pilot_v2_claims_by_source.items()
        )
    ),
)

print(
    "Claims by authority role:",
    dict(
        sorted(
            pilot_v2_claims_by_role.items()
        )
    ),
)

print(
    "Claims by type:",
    dict(
        sorted(
            pilot_v2_claims_by_type.items()
        )
    ),
)


print()
print(
    "Unresolved Window failures:",
    len(
        pilot_v2_failures
    ),
)

print(
    "Duplicate Raw Claim IDs:",
    (
        len(
            pilot_v2_raw_claim_ids
        )
        - len(
            set(
                pilot_v2_raw_claim_ids
            )
        )
    ),
)

print(
    "Duplicate Evidence IDs:",
    (
        len(
            pilot_v2_evidence_ids
        )
        - len(
            set(
                pilot_v2_evidence_ids
            )
        )
    ),
)

print(
    "Structural grounding failures:",
    0,
)


print()
print(
    "Pilot v1 Semantic decisions:",
    baseline_semantic_report[
        "decision_counts"
    ],
)


print()
print(
    "=== SAMPLE PILOT V2 CLAIMS ==="
)


pilot_v2_evidence_by_claim = defaultdict(
    list
)


for evidence in (
    pilot_v2_evidence_links
):

    pilot_v2_evidence_by_claim[
        evidence[
            "raw_claim_id"
        ]
    ].append(
        evidence
    )


for claim in (
    pilot_v2_raw_claims[:15]
):

    print()
    print(
        "Raw Claim ID:",
        claim[
            "raw_claim_id"
        ],
    )

    print(
        "Type:",
        claim[
            "claim_type"
        ],
    )

    print(
        "Claim:",
        claim[
            "claim_text_fa"
        ],
    )


    for evidence in (
        pilot_v2_evidence_by_claim[
            claim[
                "raw_claim_id"
            ]
        ]
    ):

        quote = (
            evidence[
                "evidence_quote_exact"
            ]
        )


        if len(
            quote
        ) > 220:

            quote = (
                quote[:217]
                + "..."
            )


        print(
            "Evidence:",
            quote,
        )


print()
print(
    "Raw Artifact:",
    pilot_v2_raw_path.relative_to(
        project_root
    ),
)

print(
    "Raw SHA-256:",
    pilot_v2_raw_sha256,
)

print(
    "Raw write status:",
    pilot_v2_raw_write_status,
)


print()
print(
    "Evidence Artifact:",
    pilot_v2_evidence_path.relative_to(
        project_root
    ),
)

print(
    "Evidence SHA-256:",
    pilot_v2_evidence_sha256,
)

print(
    "Evidence write status:",
    pilot_v2_evidence_write_status,
)


print()
print(
    "Report:",
    pilot_v2_report_path.relative_to(
        project_root
    ),
)

print(
    "Report write status:",
    pilot_v2_report_write_status,
)


print()
print(
    "Semantic Quality Gate v2:",
    "NOT_YET_RUN",
)

print(
    "Full Registry Extraction:",
    "NOT_STARTED",
)


print()
print(
    "CODE CELL 12 GATE:",
    pilot_v2_extraction_status,
)
```

<h2>Semantic Quality Gate v2 و تصمیم Freeze یا Repair دوباره</h2>

<p>
Prompt Extraction v2 دقیقاً روی همان ۱۵ Prompt Window قبلی اجرا شد و ۵۷ Claim با Grounding ساختاری کامل تولید کرد.
تعداد Claim نسبت به Pilot v1 کمی کاهش یافته است، اما مشاهدهٔ نمونه‌های واقعی نشان می‌دهد بعضی Fragmentها، گزاره‌های وابسته به Context و List Headerها هنوز از مرحلهٔ Extraction عبور کرده‌اند.
</p>

<p>
بنابراین در این مرحله Semantic Gate نیز سخت‌گیرانه‌تر می‌شود.
هدف فقط شمارش ACCEPT و REJECT نیست؛ هدف تعیین این است که آیا ترکیب Extraction v2 + Semantic Validation می‌تواند Contract قابل اتکایی برای Full Registry باشد یا هنوز باید Prompt Extraction دوباره Repair شود.
</p>

<h3>تغییر مهم نسبت به Semantic Gate v1</h3>

<p>
Surrounding Context فقط برای فهم Source و تشخیص وابستگی Claim استفاده می‌شود.
این Context اجازه ندارد Evidence جدیدی برای Repair ایجاد کند.
</p>

<p>
اگر Claim فقط با اضافه‌کردن Factی از Context اطراف قابل اصلاح باشد ولی آن Fact در Evidence Linkهای فعلی وجود نداشته باشد، تصمیم باید <code>REJECT</code> باشد.
</p>

<h3>معیارهای Semantic Gate v2</h3>

<ul>
<li><strong>Entailment:</strong> کل Claim باید از Evidenceهای ثبت‌شده نتیجه شود.</li>
<li><strong>Evidence Completeness:</strong> تمام اجزای معنایی Claim باید در Evidence موجود باشند.</li>
<li><strong>Standalone:</strong> Claim باید بدون Context بیرونی قابل فهم باشد.</li>
<li><strong>Atomicity:</strong> Claim باید یک واحد دانش قابل مدیریت باشد.</li>
<li><strong>Domain Relevance:</strong> Claim باید دانش مالی، اقتصادی یا مالی شخصی باشد.</li>
<li><strong>Knowledge Utility:</strong> Claim باید چیزی بیشتر از Fragment، List Header، Source Commentary یا گزارهٔ کم‌محتوا باشد.</li>
<li><strong>Claim Type Fit:</strong> نوع Claim باید با معنای واقعی گزاره سازگار باشد.</li>
</ul>

<h3>Failureهای مهمی که باید صریحاً تشخیص داده شوند</h3>

<ul>
<li>فاعل یا موضوع نامشخص؛</li>
<li>شروع Fragment با عباراتی مانند «که»، «دلیل عمده‌ش»، «این رو»، «بهش» یا مرجع نامشخص؛</li>
<li>عدد بدون موضوع یا معنای کامل؛</li>
<li>List Header بدون خود آیتم‌ها؛</li>
<li>Source Metadata یا Editorial Commentary؛</li>
<li>توصیف کتاب، پرسشنامه، نسخه یا شیوهٔ نگارش به‌جای دانش دامنه؛</li>
<li>Claimی که تنها با Context خارج از Evidence قابل تکمیل است؛</li>
<li>Claim کم‌ارزش یا بسیار موقعیتی که به‌عنوان واحد پایدار Registry مناسب نیست.</li>
</ul>

<h3>Repair Contract سخت‌گیرانه</h3>

<p>
Repair فقط زمانی مجاز است که تمام Factهای Claim اصلاح‌شده در همان Evidence Linkهای موجود حاضر باشند.
Evidence جدید در Semantic Gate ساخته نمی‌شود.
</p>

<p>
Repair مناسب می‌تواند شامل مواردی مانند اصلاح Claim Type، حذف یک عبارت زائد یا بازنویسی محدود برای Standalone شدن باشد؛ اما افزودن فاعل، عدد، رابطه یا Fact جدیدی که Evidence فعلی آن را نشان نمی‌دهد ممنوع است.
</p>

<h3>Risk Signalهای Deterministic</h3>

<p>
برای کمک به Judge، چند Signal صرفاً تشخیصی از متن Claim محاسبه می‌شود؛ مانند شروع Fragment، عبارت‌های ارجاعی، Numeric Fragment و List-header.
این Signalها به‌تنهایی تصمیم نهایی را تعیین نمی‌کنند، اما باعث می‌شوند Judge Failureهای مشاهده‌شده در Pilot را نادیده نگیرد.
</p>

<h3>Artifactهای این مرحله</h3>

<ul>
<li><code>claim_extraction_pilot_semantic_adjudication_v2.jsonl</code></li>
<li><code>claim_extraction_pilot_semantic_report_v2.json</code></li>
<li><code>claim_extraction_pilot_semantic_cache_v2/</code></li>
</ul>

<h3>اصل مهم</h3>

<p>
این مرحله Full Corpus را اجرا نمی‌کند و به‌صورت خودکار Prompt را Freeze نمی‌کند.
پس از مشاهدهٔ Decisionها، Issueها و نمونه‌های واقعی، تصمیم Freeze یا Repair Again گرفته خواهد شد.
</p>

<h3>معیار PASS</h3>

<ul>
<li>تمام Claimهای Pilot v2 دقیقاً یک Judgment داشته باشند؛</li>
<li>تمام Decisionها و Checkها با Contract سازگار باشند؛</li>
<li>Repair فقط به Evidenceهای موجود متکی باشد؛</li>
<li>هیچ Claim یا Evidence جدید در Semantic Gate ساخته نشود؛</li>
<li>هیچ Batch Failure حل‌نشده باقی نماند؛</li>
<li>Artifactها بدون Overwrite ناامن منتشر شوند؛</li>
<li>Artifactها پس از Publication دوباره از Disk Validate شوند؛</li>
<li>Full Registry Extraction همچنان شروع نشود.</li>
</ul>

</div>

```python
# Code Cell 13

assert (
    pilot_v2_extraction_status
    == "PASS"
), (
    "Code Cell 12 must PASS before "
    "running Semantic Quality Gate v2."
)


SEMANTIC_V2_PROMPT_VERSION = (
    "claim_semantic_adjudication_v2"
)

SEMANTIC_V2_BATCH_SIZE = 4
SEMANTIC_V2_MAX_TOKENS = 4000
SEMANTIC_V2_TEMPERATURE = 0
SEMANTIC_V2_MAX_ATTEMPTS = 3
SEMANTIC_V2_CONTEXT_CHARS = 350


SEMANTIC_V2_CHECK_NAMES = {
    "entailment",
    "evidence_completeness",
    "standalone",
    "atomicity",
    "domain_relevance",
    "knowledge_utility",
    "claim_type_fit",
}


SEMANTIC_V2_DECISIONS = {
    "ACCEPT",
    "REPAIR",
    "REJECT",
}


SEMANTIC_V2_ISSUES = {
    "UNSUPPORTED_OR_OVERSTATED",
    "EVIDENCE_INCOMPLETE",
    "CONTEXT_DEPENDENT",
    "SUBJECT_OR_REFERENT_UNRESOLVED",
    "NOT_ATOMIC",
    "CLAIM_TYPE_MISMATCH",
    "SOURCE_METADATA_OR_COMMENTARY",
    "NON_DOMAIN_KNOWLEDGE",
    "LIST_HEADER_INCOMPLETE",
    "NUMERIC_SUBJECT_MISSING",
    "IMPERATIVE_OR_FRAGMENT",
    "TOO_VAGUE",
    "LOW_KNOWLEDGE_UTILITY",
    "OTHER",
}


semantic_v2_adjudication_path = (
    registry_artifact_root
    / "claim_extraction_pilot_semantic_adjudication_v2.jsonl"
)

semantic_v2_report_path = (
    registry_artifact_root
    / "claim_extraction_pilot_semantic_report_v2.json"
)

semantic_v2_cache_root = (
    registry_artifact_root
    / "claim_extraction_pilot_semantic_cache_v2"
)


# ---------------------------------------------------------
# Independent reload of Pilot v2 artifacts.
# ---------------------------------------------------------

semantic_v2_claims = load_jsonl(
    pilot_v2_raw_path
)

semantic_v2_evidence = load_jsonl(
    pilot_v2_evidence_path
)

semantic_v2_source_report = load_json(
    pilot_v2_report_path
)


assert (
    len(
        semantic_v2_claims
    )
    == semantic_v2_source_report[
        "results"
    ][
        "raw_claims"
    ]
)


assert (
    len(
        semantic_v2_evidence
    )
    == semantic_v2_source_report[
        "results"
    ][
        "evidence_links"
    ]
)


semantic_v2_claim_by_id = {
    item[
        "raw_claim_id"
    ]: item
    for item in (
        semantic_v2_claims
    )
}


assert (
    len(
        semantic_v2_claim_by_id
    )
    == len(
        semantic_v2_claims
    )
)


semantic_v2_evidence_by_claim = defaultdict(
    list
)


for evidence in (
    semantic_v2_evidence
):

    raw_claim_id = evidence[
        "raw_claim_id"
    ]

    assert (
        raw_claim_id
        in semantic_v2_claim_by_id
    )

    semantic_v2_evidence_by_claim[
        raw_claim_id
    ].append(
        evidence
    )


for raw_claim_id in (
    semantic_v2_claim_by_id
):

    assert (
        semantic_v2_evidence_by_claim[
            raw_claim_id
        ]
    )


# ---------------------------------------------------------
# Deterministic risk signals.
# Signals inform the semantic judge;
# they do not directly decide ACCEPT/REJECT.
# ---------------------------------------------------------

def detect_claim_risk_signals(
    claim_text,
):
    normalized = (
        " ".join(
            str(
                claim_text
            ).strip().split()
        )
    )

    signals = []


    fragment_starts = [
        "که ",
        "و ",
        "اما ",
        "بنابراین ",
        "پس ",
        "دلیل عمده‌ش",
        "دلیل عمده اش",
        "می‌گوید که ",
        "میگه که ",
        "این رو ",
        "اون رو ",
        "بهش ",
    ]


    lowered = (
        normalized.casefold()
    )


    for prefix in fragment_starts:

        if lowered.startswith(
            prefix.casefold()
        ):

            signals.append(
                "FRAGMENT_OR_DISCOURSE_START"
            )

            break


    unresolved_markers = [
        " این ",
        " آن ",
        " اون ",
        " اینو ",
        " اونو ",
        " بهش ",
        " خودش ",
        " این‌ها ",
        " آنها ",
    ]


    padded = (
        f" {lowered} "
    )


    if any(
        marker.casefold()
        in padded
        for marker in (
            unresolved_markers
        )
    ):

        signals.append(
            "POSSIBLE_UNRESOLVED_REFERENT"
        )


    digit_or_percent = bool(
        re.search(
            r"[\d۰-۹٪%]",
            normalized,
        )
    )


    short_numeric = (
        digit_or_percent
        and len(
            normalized
        ) < 70
    )


    if short_numeric:
        signals.append(
            "SHORT_NUMERIC_CLAIM"
        )


    count_header_patterns = [
        r"\bدو\s+.+\s+وجود\s+دارد",
        r"\bسه\s+.+\s+وجود\s+دارد",
        r"\bچند\s+.+\s+وجود\s+دارد",
        r"\bدو\s+.+\s+عبارت",
        r"\bسه\s+.+\s+عبارت",
    ]


    if any(
        re.search(
            pattern,
            normalized,
        )
        for pattern in (
            count_header_patterns
        )
    ):

        signals.append(
            "POSSIBLE_LIST_HEADER"
        )


    source_commentary_terms = [
        "این کتاب",
        "این پرسشنامه",
        "این فصل",
        "این نسخه",
        "نویسنده در این",
        "در کتاب حاضر",
        "برای مخاطب ایرانی",
        "ویرایش",
    ]


    if any(
        term in normalized
        for term in (
            source_commentary_terms
        )
    ):

        signals.append(
            "POSSIBLE_SOURCE_COMMENTARY"
        )


    return sorted(
        set(
            signals
        )
    )


semantic_v2_risk_signal_counts = Counter()


for claim in (
    semantic_v2_claims
):

    signals = (
        detect_claim_risk_signals(
            claim[
                "claim_text_fa"
            ]
        )
    )


    for signal in signals:
        semantic_v2_risk_signal_counts[
            signal
        ] += 1


# ---------------------------------------------------------
# Materialize Evidence + limited surrounding context.
# Surrounding context is judge-only, never new Evidence.
# ---------------------------------------------------------

def semantic_v2_context_for_evidence(
    evidence,
):
    document_id = evidence[
        "document_id"
    ]

    source_text = (
        accepted_records_by_id[
            document_id
        ][
            canonical_text_field
        ]
    )

    evidence_start = (
        evidence[
            "locator"
        ][
            "text_char_start"
        ]
    )

    evidence_end = (
        evidence[
            "locator"
        ][
            "text_char_end"
        ]
    )

    quote = evidence[
        "evidence_quote_exact"
    ]


    assert (
        source_text[
            evidence_start:
            evidence_end
        ]
        == quote
    )


    context_start = max(
        0,
        evidence_start
        - SEMANTIC_V2_CONTEXT_CHARS,
    )

    context_end = min(
        len(
            source_text
        ),
        evidence_end
        + SEMANTIC_V2_CONTEXT_CHARS,
    )


    return {
        "evidence_id": (
            evidence[
                "evidence_id"
            ]
        ),
        "document_id": (
            document_id
        ),
        "source_type": (
            evidence[
                "source_type"
            ]
        ),
        "quote_exact": (
            quote
        ),
        "judge_only_surrounding_context": (
            source_text[
                context_start:
                context_end
            ]
        ),
    }


semantic_v2_items = []


for claim in sorted(
    semantic_v2_claims,
    key=lambda item: (
        item[
            "raw_claim_id"
        ]
    ),
):

    raw_claim_id = claim[
        "raw_claim_id"
    ]


    semantic_v2_items.append({
        "raw_claim_id": (
            raw_claim_id
        ),
        "claim_text_fa": (
            claim[
                "claim_text_fa"
            ]
        ),
        "claim_type": (
            claim[
                "claim_type"
            ]
        ),
        "authority_class": (
            claim[
                "authority_class"
            ]
        ),
        "claim_discovery_role": (
            claim[
                "claim_discovery_role"
            ]
        ),
        "risk_signals": (
            detect_claim_risk_signals(
                claim[
                    "claim_text_fa"
                ]
            )
        ),
        "evidence": [
            semantic_v2_context_for_evidence(
                evidence
            )
            for evidence in (
                semantic_v2_evidence_by_claim[
                    raw_claim_id
                ]
            )
        ],
    })


semantic_v2_batches = [
    semantic_v2_items[
        start:
        start
        + SEMANTIC_V2_BATCH_SIZE
    ]
    for start in range(
        0,
        len(
            semantic_v2_items
        ),
        SEMANTIC_V2_BATCH_SIZE,
    )
]


def semantic_v2_batch_id(
    batch_items,
):
    material = stable_json({
        "prompt_version": (
            SEMANTIC_V2_PROMPT_VERSION
        ),
        "model": (
            pilot_chat_model
        ),
        "claim_ids": [
            item[
                "raw_claim_id"
            ]
            for item in (
                batch_items
            )
        ],
    })


    digest = hashlib.sha256(
        material.encode(
            "utf-8"
        )
    ).hexdigest()[:24]


    return (
        f"semantic-v2-batch:{digest}"
    )


# ---------------------------------------------------------
# Semantic Judge v2 prompt.
# ---------------------------------------------------------

def build_semantic_v2_prompt(
    batch_items,
    retry_feedback=None,
):
    retry_block = ""


    if retry_feedback:

        retry_block = f"""

Your previous response failed deterministic validation.

Validator error:
{retry_feedback}

Return the COMPLETE corrected JSON for EVERY raw_claim_id.
"""


    system_message = """
You are a strict adjudicator for a high-quality Persian financial knowledge registry.

You are NOT extracting new claims.
You are judging already extracted claims.

Use only:
- the Claim,
- its registered exact Evidence quotes,
- limited surrounding context provided only for understanding the source.

CRITICAL EVIDENCE RULE:

Surrounding context is JUDGE-ONLY context.
It is NOT registered Evidence.

You may use surrounding context to recognize that a Claim is incomplete or context-dependent.
You may NOT use surrounding context to add a new fact during REPAIR.

A repaired Claim must be fully supported by the EXISTING exact Evidence quotes alone.

If a Claim needs a subject, referent, number meaning, list item, causal component, or other factual content that exists only in surrounding context and not in the registered Evidence, choose REJECT.

Registry quality standard:

A valid registry Claim must:
1. be fully entailed by registered Evidence,
2. have complete Evidence for all meaningful parts,
3. be understandable standalone,
4. be atomic enough,
5. represent financial/economic/personal-finance domain knowledge,
6. have durable knowledge utility,
7. use the correct claim type.

Reject:
- conversational fragments,
- unresolved "this/that/he/it" references,
- statements beginning mid-thought,
- numeric fragments without an explicit subject,
- list headers without the substantive listed knowledge,
- source/editorial/book/questionnaire commentary,
- statements that are merely about the wording or construction of the source,
- vague situation-specific fragments,
- claims whose full meaning is not supported by registered Evidence,
- low-information claims unsuitable as independent knowledge units.

REPAIR is conservative:
- it may fix wording or claim_type,
- it may remove unsupported excess wording,
- it may make a Claim clearer only when the needed facts already appear in the registered Evidence,
- it may NOT import facts from surrounding context.

Return valid JSON only.
No markdown.
No text outside JSON.
""".strip()


    user_message = f"""
Adjudicate every Claim.

Allowed decisions:
{json.dumps(sorted(SEMANTIC_V2_DECISIONS), ensure_ascii=False)}

Required checks:
{json.dumps(sorted(SEMANTIC_V2_CHECK_NAMES), ensure_ascii=False)}

Each check value must be PASS or FAIL.

Allowed issues:
{json.dumps(sorted(SEMANTIC_V2_ISSUES), ensure_ascii=False)}

Allowed claim types:
{json.dumps(sorted(ALLOWED_CLAIM_TYPES), ensure_ascii=False)}

Required schema:

{{
  "judgments": [
    {{
      "raw_claim_id": "raw-claim:...",
      "decision": "ACCEPT | REPAIR | REJECT",
      "checks": {{
        "entailment": "PASS | FAIL",
        "evidence_completeness": "PASS | FAIL",
        "standalone": "PASS | FAIL",
        "atomicity": "PASS | FAIL",
        "domain_relevance": "PASS | FAIL",
        "knowledge_utility": "PASS | FAIL",
        "claim_type_fit": "PASS | FAIL"
      }},
      "issues": [
        "CONTROLLED_ISSUE"
      ],
      "repaired_claim_text_fa": null,
      "recommended_claim_type": null,
      "repair_basis_evidence_ids": [],
      "brief_reason_fa": "دلیل کوتاه و دقیق"
    }}
  ]
}}

Decision consistency:

ACCEPT:
- every check = PASS,
- repaired_claim_text_fa = null,
- recommended_claim_type = original claim_type,
- repair_basis_evidence_ids = [].

REPAIR:
- entailment = PASS,
- evidence_completeness = PASS,
- domain_relevance = PASS,
- knowledge_utility = PASS,
- repaired_claim_text_fa must be non-empty,
- recommended_claim_type must be an allowed type,
- repair_basis_evidence_ids must contain one or more EXISTING evidence_ids,
- repaired Claim must use only facts stated in those registered Evidence quotes.

REJECT:
- repaired_claim_text_fa = null,
- recommended_claim_type = null,
- repair_basis_evidence_ids = [].

Risk signals are warnings only.
Independently inspect the Claim and Evidence.

Items:

{json.dumps(batch_items, ensure_ascii=False, indent=2)}

{retry_block}
""".strip()


    return (
        system_message,
        user_message,
    )


def call_semantic_v2_judge(
    system_message,
    user_message,
):
    response = (
        pilot_client
        .chat
        .completions
        .create(
            model=(
                pilot_chat_model
            ),
            messages=[
                {
                    "role": "system",
                    "content": (
                        system_message
                    ),
                },
                {
                    "role": "user",
                    "content": (
                        user_message
                    ),
                },
            ],
            temperature=(
                SEMANTIC_V2_TEMPERATURE
            ),
            max_tokens=(
                SEMANTIC_V2_MAX_TOKENS
            ),
        )
    )


    content = (
        response
        .choices[0]
        .message
        .content
    )


    if not isinstance(
        content,
        str,
    ):
        raise ValueError(
            "Semantic v2 response "
            "content is not a string."
        )


    return content


# ---------------------------------------------------------
# Deterministic response validation.
# ---------------------------------------------------------

def validate_semantic_v2_payload(
    batch_items,
    payload,
):
    if not isinstance(
        payload,
        dict,
    ):
        raise ValueError(
            "Semantic v2 payload "
            "must be an object."
        )


    judgments = payload.get(
        "judgments"
    )


    if not isinstance(
        judgments,
        list,
    ):
        raise ValueError(
            "'judgments' must be a list."
        )


    expected_ids = {
        item[
            "raw_claim_id"
        ]
        for item in (
            batch_items
        )
    }


    original_by_id = {
        item[
            "raw_claim_id"
        ]: item
        for item in (
            batch_items
        )
    }


    observed_ids = [
        item.get(
            "raw_claim_id"
        )
        if isinstance(
            item,
            dict,
        )
        else None
        for item in (
            judgments
        )
    ]


    if (
        len(
            observed_ids
        )
        != len(
            set(
                observed_ids
            )
        )
    ):
        raise ValueError(
            "Duplicate raw_claim_id "
            "in Semantic v2 response."
        )


    if (
        set(
            observed_ids
        )
        != expected_ids
    ):
        raise ValueError(
            "Semantic v2 Claim IDs "
            "do not exactly match batch."
        )


    validated = []


    for judgment in judgments:

        raw_claim_id = judgment[
            "raw_claim_id"
        ]

        original = (
            original_by_id[
                raw_claim_id
            ]
        )


        decision = str(
            judgment.get(
                "decision",
                "",
            )
        ).strip()


        if (
            decision
            not in SEMANTIC_V2_DECISIONS
        ):
            raise ValueError(
                "Invalid Semantic v2 decision: "
                f"{decision!r}"
            )


        checks = judgment.get(
            "checks"
        )


        if not isinstance(
            checks,
            dict,
        ):
            raise ValueError(
                "Semantic v2 checks "
                "must be an object."
            )


        if (
            set(
                checks
            )
            != SEMANTIC_V2_CHECK_NAMES
        ):
            raise ValueError(
                "Semantic v2 check keys "
                "do not match Contract."
            )


        for (
            check_name,
            check_value,
        ) in checks.items():

            if (
                check_value
                not in {
                    "PASS",
                    "FAIL",
                }
            ):
                raise ValueError(
                    f"Invalid {check_name} "
                    f"value: {check_value!r}"
                )


        issues = judgment.get(
            "issues",
            []
        )


        if not isinstance(
            issues,
            list,
        ):
            raise ValueError(
                "Issues must be a list."
            )


        if (
            len(
                issues
            )
            != len(
                set(
                    issues
                )
            )
        ):
            raise ValueError(
                "Duplicate Semantic v2 issue."
            )


        unknown_issues = (
            set(
                issues
            )
            - SEMANTIC_V2_ISSUES
        )


        if unknown_issues:
            raise ValueError(
                "Unknown Semantic v2 issues: "
                f"{sorted(unknown_issues)}"
            )


        repaired_claim = judgment.get(
            "repaired_claim_text_fa"
        )

        recommended_type = judgment.get(
            "recommended_claim_type"
        )

        repair_basis = judgment.get(
            "repair_basis_evidence_ids",
            [],
        )


        if not isinstance(
            repair_basis,
            list,
        ):
            raise ValueError(
                "repair_basis_evidence_ids "
                "must be a list."
            )


        allowed_evidence_ids = {
            item[
                "evidence_id"
            ]
            for item in (
                original[
                    "evidence"
                ]
            )
        }


        if not set(
            repair_basis
        ).issubset(
            allowed_evidence_ids
        ):
            raise ValueError(
                "Repair references Evidence "
                "outside the Claim."
            )


        brief_reason = str(
            judgment.get(
                "brief_reason_fa",
                "",
            )
        ).strip()


        if not brief_reason:
            raise ValueError(
                "brief_reason_fa is empty."
            )


        if decision == "ACCEPT":

            if any(
                value != "PASS"
                for value in (
                    checks.values()
                )
            ):
                raise ValueError(
                    "ACCEPT requires "
                    "all checks PASS."
                )


            if repaired_claim is not None:
                raise ValueError(
                    "ACCEPT may not include "
                    "a repaired Claim."
                )


            if (
                recommended_type
                != original[
                    "claim_type"
                ]
            ):
                raise ValueError(
                    "ACCEPT must preserve "
                    "original Claim Type."
                )


            if repair_basis:
                raise ValueError(
                    "ACCEPT must have "
                    "empty repair basis."
                )


        elif decision == "REPAIR":

            for required_pass in [
                "entailment",
                "evidence_completeness",
                "domain_relevance",
                "knowledge_utility",
            ]:

                if (
                    checks[
                        required_pass
                    ]
                    != "PASS"
                ):
                    raise ValueError(
                        "REPAIR requires "
                        f"{required_pass}=PASS."
                    )


            if (
                not isinstance(
                    repaired_claim,
                    str,
                )
                or not repaired_claim.strip()
            ):
                raise ValueError(
                    "REPAIR requires a "
                    "non-empty repaired Claim."
                )


            repaired_claim = (
                repaired_claim.strip()
            )


            if (
                recommended_type
                not in ALLOWED_CLAIM_TYPES
            ):
                raise ValueError(
                    "REPAIR has invalid "
                    "recommended Claim Type."
                )


            if not repair_basis:
                raise ValueError(
                    "REPAIR requires at least "
                    "one existing Evidence ID."
                )


        else:

            if repaired_claim is not None:
                raise ValueError(
                    "REJECT must not include "
                    "a repaired Claim."
                )


            if recommended_type is not None:
                raise ValueError(
                    "REJECT must use "
                    "recommended_claim_type=null."
                )


            if repair_basis:
                raise ValueError(
                    "REJECT must have "
                    "empty repair basis."
                )


        if (
            "EVIDENCE_INCOMPLETE"
            in issues
            and checks[
                "evidence_completeness"
            ]
            != "FAIL"
        ):
            raise ValueError(
                "EVIDENCE_INCOMPLETE requires "
                "evidence_completeness=FAIL."
            )


        if (
            "UNSUPPORTED_OR_OVERSTATED"
            in issues
            and checks[
                "entailment"
            ]
            != "FAIL"
        ):
            raise ValueError(
                "UNSUPPORTED_OR_OVERSTATED "
                "requires entailment=FAIL."
            )


        if (
            "CLAIM_TYPE_MISMATCH"
            in issues
            and checks[
                "claim_type_fit"
            ]
            != "FAIL"
        ):
            raise ValueError(
                "CLAIM_TYPE_MISMATCH requires "
                "claim_type_fit=FAIL."
            )


        if (
            "NON_DOMAIN_KNOWLEDGE"
            in issues
            and checks[
                "domain_relevance"
            ]
            != "FAIL"
        ):
            raise ValueError(
                "NON_DOMAIN_KNOWLEDGE requires "
                "domain_relevance=FAIL."
            )


        if (
            "LOW_KNOWLEDGE_UTILITY"
            in issues
            and checks[
                "knowledge_utility"
            ]
            != "FAIL"
        ):
            raise ValueError(
                "LOW_KNOWLEDGE_UTILITY requires "
                "knowledge_utility=FAIL."
            )


        validated.append({
            "raw_claim_id": (
                raw_claim_id
            ),
            "decision": (
                decision
            ),
            "checks": {
                key: checks[
                    key
                ]
                for key in sorted(
                    SEMANTIC_V2_CHECK_NAMES
                )
            },
            "issues": sorted(
                issues
            ),
            "risk_signals": (
                original[
                    "risk_signals"
                ]
            ),
            "repaired_claim_text_fa": (
                repaired_claim
            ),
            "recommended_claim_type": (
                recommended_type
            ),
            "repair_basis_evidence_ids": (
                sorted(
                    repair_basis
                )
            ),
            "brief_reason_fa": (
                brief_reason
            ),
            "original_claim_text_fa": (
                original[
                    "claim_text_fa"
                ]
            ),
            "original_claim_type": (
                original[
                    "claim_type"
                ]
            ),
            "semantic_model": (
                pilot_chat_model
            ),
            "semantic_prompt_version": (
                SEMANTIC_V2_PROMPT_VERSION
            ),
        })


    return sorted(
        validated,
        key=lambda item: (
            item[
                "raw_claim_id"
            ]
        ),
    )


# ---------------------------------------------------------
# Execute batches with validated cache.
# ---------------------------------------------------------

semantic_v2_cache_root.mkdir(
    parents=True,
    exist_ok=True,
)


semantic_v2_judgments = []

semantic_v2_failures = []

semantic_v2_api_calls_this_run = 0
semantic_v2_cache_hits_this_run = 0


print(
    "=== NB12i PILOT SEMANTIC QUALITY GATE V2 ==="
)

print()
print(
    "Pilot v2 Claims:",
    len(
        semantic_v2_items
    ),
)

print(
    "Semantic v2 batches:",
    len(
        semantic_v2_batches
    ),
)

print(
    "Batch size:",
    SEMANTIC_V2_BATCH_SIZE,
)

print()


for batch_number, batch_items in enumerate(
    semantic_v2_batches,
    1,
):

    batch_id = (
        semantic_v2_batch_id(
            batch_items
        )
    )


    cache_path = (
        semantic_v2_cache_root
        / (
            batch_id.replace(
                ":",
                "_",
            )
            + ".json"
        )
    )


    print(
        f"[{batch_number:02d}/{len(semantic_v2_batches):02d}]",
        batch_id,
        "| claims:",
        len(
            batch_items
        ),
    )


    if cache_path.exists():

        cached = load_json(
            cache_path
        )


        assert (
            cached[
                "batch_id"
            ]
            == batch_id
        )


        assert (
            cached[
                "semantic_prompt_version"
            ]
            == SEMANTIC_V2_PROMPT_VERSION
        )


        assert (
            cached[
                "model"
            ]
            == pilot_chat_model
        )


        cached_judgments = (
            cached[
                "judgments"
            ]
        )


        assert (
            {
                item[
                    "raw_claim_id"
                ]
                for item in (
                    cached_judgments
                )
            }
            == {
                item[
                    "raw_claim_id"
                ]
                for item in (
                    batch_items
                )
            }
        )


        semantic_v2_judgments.extend(
            cached_judgments
        )


        semantic_v2_cache_hits_this_run += 1


        print(
            "   cache: HIT"
        )


        continue


    validated_judgments = None
    last_error = None


    for attempt in range(
        1,
        SEMANTIC_V2_MAX_ATTEMPTS + 1,
    ):

        (
            system_message,
            user_message,
        ) = build_semantic_v2_prompt(
            batch_items,
            retry_feedback=(
                last_error
            ),
        )


        try:

            content = (
                call_semantic_v2_judge(
                    system_message,
                    user_message,
                )
            )


            semantic_v2_api_calls_this_run += 1


            payload = parse_model_json(
                content
            )


            validated_judgments = (
                validate_semantic_v2_payload(
                    batch_items,
                    payload,
                )
            )


            break


        except Exception as exc:

            last_error = (
                f"{type(exc).__name__}: "
                f"{exc}"
            )


            print(
                f"   attempt {attempt}:",
                "INVALID",
                "|",
                last_error,
            )


            if (
                attempt
                < SEMANTIC_V2_MAX_ATTEMPTS
            ):
                time.sleep(
                    min(
                        2.0,
                        float(
                            attempt
                        ),
                    )
                )


    if (
        validated_judgments
        is None
    ):

        semantic_v2_failures.append({
            "batch_id": (
                batch_id
            ),
            "raw_claim_ids": [
                item[
                    "raw_claim_id"
                ]
                for item in (
                    batch_items
                )
            ],
            "last_error": (
                last_error
            ),
        })


        print(
            "   result: FAIL"
        )


        continue


    cache_payload = {
        "batch_id": (
            batch_id
        ),
        "semantic_prompt_version": (
            SEMANTIC_V2_PROMPT_VERSION
        ),
        "model": (
            pilot_chat_model
        ),
        "judgments": (
            validated_judgments
        ),
    }


    cache_text = (
        json.dumps(
            cache_payload,
            ensure_ascii=False,
            indent=2,
            sort_keys=True,
        )
        + "\n"
    )


    with cache_path.open(
        "x",
        encoding="utf-8",
    ) as file:
        file.write(
            cache_text
        )


    semantic_v2_judgments.extend(
        validated_judgments
    )


    print(
        "   result: VALID"
    )


print()


assert not semantic_v2_failures, (
    "Semantic v2 has unresolved failures:\n"
    + json.dumps(
        semantic_v2_failures,
        ensure_ascii=False,
        indent=2,
    )
)


semantic_v2_judgments = sorted(
    semantic_v2_judgments,
    key=lambda item: (
        item[
            "raw_claim_id"
        ]
    ),
)


semantic_v2_judged_ids = [
    item[
        "raw_claim_id"
    ]
    for item in (
        semantic_v2_judgments
    )
]


assert (
    len(
        semantic_v2_judged_ids
    )
    == len(
        semantic_v2_claims
    )
)


assert (
    len(
        semantic_v2_judged_ids
    )
    == len(
        set(
            semantic_v2_judged_ids
        )
    )
)


assert (
    set(
        semantic_v2_judged_ids
    )
    == set(
        semantic_v2_claim_by_id
    )
)


# ---------------------------------------------------------
# Aggregate results.
# ---------------------------------------------------------

semantic_v2_decision_counts = Counter(
    item[
        "decision"
    ]
    for item in (
        semantic_v2_judgments
    )
)


semantic_v2_issue_counts = Counter()

semantic_v2_failed_check_counts = Counter()


for item in (
    semantic_v2_judgments
):

    for issue in (
        item[
            "issues"
        ]
    ):
        semantic_v2_issue_counts[
            issue
        ] += 1


    for (
        check_name,
        check_value,
    ) in (
        item[
            "checks"
        ].items()
    ):

        if (
            check_value
            == "FAIL"
        ):
            semantic_v2_failed_check_counts[
                check_name
            ] += 1


semantic_v2_decision_by_source = defaultdict(
    Counter
)

semantic_v2_decision_by_role = defaultdict(
    Counter
)


for judgment in (
    semantic_v2_judgments
):

    claim = (
        semantic_v2_claim_by_id[
            judgment[
                "raw_claim_id"
            ]
        ]
    )

    window = (
        pilot_v2_window_by_id[
            claim[
                "prompt_window_id"
            ]
        ]
    )


    semantic_v2_decision_by_source[
        window[
            "source_type"
        ]
    ][
        judgment[
            "decision"
        ]
    ] += 1


    semantic_v2_decision_by_role[
        claim[
            "claim_discovery_role"
        ]
    ][
        judgment[
            "decision"
        ]
    ] += 1


# ---------------------------------------------------------
# Publish v2 Semantic artifacts.
# ---------------------------------------------------------

semantic_v2_adjudication_text = "".join(
    json.dumps(
        item,
        ensure_ascii=False,
        sort_keys=True,
    )
    + "\n"
    for item in (
        semantic_v2_judgments
    )
)


semantic_v2_adjudication_sha256 = (
    hashlib.sha256(
        semantic_v2_adjudication_text.encode(
            "utf-8"
        )
    ).hexdigest()
)


semantic_v2_report = {
    "report_version": (
        "claim_extraction_pilot_semantic_report_v2"
    ),

    "source_pilot_report": (
        pilot_v2_report_path.name
    ),

    "source_raw_claims": (
        pilot_v2_raw_path.name
    ),

    "source_raw_claims_sha256": (
        file_sha256(
            pilot_v2_raw_path
        )
    ),

    "semantic_prompt_version": (
        SEMANTIC_V2_PROMPT_VERSION
    ),

    "semantic_model": (
        pilot_chat_model
    ),

    "claims_adjudicated": (
        len(
            semantic_v2_judgments
        )
    ),

    "decision_counts": dict(
        sorted(
            semantic_v2_decision_counts.items()
        )
    ),

    "failed_checks": dict(
        sorted(
            semantic_v2_failed_check_counts.items()
        )
    ),

    "issue_counts": dict(
        sorted(
            semantic_v2_issue_counts.items()
        )
    ),

    "risk_signal_counts": dict(
        sorted(
            semantic_v2_risk_signal_counts.items()
        )
    ),

    "decisions_by_source_type": {
        source_type: dict(
            sorted(
                counts.items()
            )
        )
        for (
            source_type,
            counts,
        ) in sorted(
            semantic_v2_decision_by_source.items()
        )
    },

    "decisions_by_claim_discovery_role": {
        role: dict(
            sorted(
                counts.items()
            )
        )
        for (
            role,
            counts,
        ) in sorted(
            semantic_v2_decision_by_role.items()
        )
    },

    "comparison_to_v1": {
        "pilot_v1_raw_claims": (
            len(
                pilot_raw_claims
            )
        ),
        "pilot_v1_semantic_decisions": (
            baseline_semantic_report[
                "decision_counts"
            ]
        ),
        "pilot_v2_raw_claims": (
            len(
                pilot_v2_raw_claims
            )
        ),
    },

    "unresolved_batch_failures": 0,

    "semantic_execution_status": (
        "PASS"
    ),

    "full_registry_readiness": (
        "AWAITING_POST_GATE_DECISION"
    ),

    "full_registry_extraction_started": (
        False
    ),

    "artifact": {
        "adjudication": (
            semantic_v2_adjudication_path.name
        ),
        "adjudication_sha256": (
            semantic_v2_adjudication_sha256
        ),
    },
}


semantic_v2_report_text = (
    json.dumps(
        semantic_v2_report,
        ensure_ascii=False,
        indent=2,
        sort_keys=True,
    )
    + "\n"
)


semantic_v2_adjudication_write_status = (
    publish_semantic_no_overwrite(
        semantic_v2_adjudication_path,
        semantic_v2_adjudication_text,
    )
)


semantic_v2_report_write_status = (
    publish_semantic_no_overwrite(
        semantic_v2_report_path,
        semantic_v2_report_text,
    )
)


disk_semantic_v2_judgments = load_jsonl(
    semantic_v2_adjudication_path
)

disk_semantic_v2_report = load_json(
    semantic_v2_report_path
)


assert (
    disk_semantic_v2_judgments
    == semantic_v2_judgments
)


assert (
    disk_semantic_v2_report
    == semantic_v2_report
)


assert (
    file_sha256(
        semantic_v2_adjudication_path
    )
    == semantic_v2_adjudication_sha256
)


pilot_semantic_v2_status = (
    "PASS"
)


# ---------------------------------------------------------
# Human-readable results.
# ---------------------------------------------------------

print(
    "=== NB12i PILOT SEMANTIC QUALITY RESULT V2 ==="
)

print()
print(
    "Claims adjudicated:",
    len(
        semantic_v2_judgments
    ),
)

print(
    "Semantic API calls this run:",
    semantic_v2_api_calls_this_run,
)

print(
    "Validated semantic cache hits:",
    semantic_v2_cache_hits_this_run,
)


print()
print(
    "Pilot v1 decisions:",
    baseline_semantic_report[
        "decision_counts"
    ],
)

print(
    "Pilot v2 decisions:",
    dict(
        sorted(
            semantic_v2_decision_counts.items()
        )
    ),
)


print()
print(
    "Failed checks:",
    dict(
        sorted(
            semantic_v2_failed_check_counts.items()
        )
    ),
)


print()
print(
    "Issue counts:",
    dict(
        sorted(
            semantic_v2_issue_counts.items()
        )
    ),
)


print()
print(
    "Deterministic risk signals:",
    dict(
        sorted(
            semantic_v2_risk_signal_counts.items()
        )
    ),
)


print()
print(
    "Decisions by source type:"
)

for (
    source_type,
    counts,
) in sorted(
    semantic_v2_decision_by_source.items()
):

    print(
        "  -",
        source_type,
        dict(
            sorted(
                counts.items()
            )
        ),
    )


print()
print(
    "Decisions by authority role:"
)

for (
    role,
    counts,
) in sorted(
    semantic_v2_decision_by_role.items()
):

    print(
        "  -",
        role,
        dict(
            sorted(
                counts.items()
            )
        ),
    )


print()
print(
    "=== V2 REPAIR CLAIMS ==="
)


semantic_v2_repairs = [
    item
    for item in (
        semantic_v2_judgments
    )
    if item[
        "decision"
    ] == "REPAIR"
]


if not semantic_v2_repairs:

    print(
        "  None"
    )


for item in (
    semantic_v2_repairs[:20]
):

    print()
    print(
        "Raw Claim ID:",
        item[
            "raw_claim_id"
        ],
    )

    print(
        "Original:",
        item[
            "original_claim_text_fa"
        ],
    )

    print(
        "Repair:",
        item[
            "repaired_claim_text_fa"
        ],
    )

    print(
        "Type:",
        item[
            "original_claim_type"
        ],
        "->",
        item[
            "recommended_claim_type"
        ],
    )

    print(
        "Issues:",
        item[
            "issues"
        ],
    )

    print(
        "Risk signals:",
        item[
            "risk_signals"
        ],
    )

    print(
        "Repair basis:",
        item[
            "repair_basis_evidence_ids"
        ],
    )

    print(
        "Reason:",
        item[
            "brief_reason_fa"
        ],
    )


if len(
    semantic_v2_repairs
) > 20:

    print(
        "  ...",
        len(
            semantic_v2_repairs
        ) - 20,
        "additional REPAIR judgments not printed",
    )


print()
print(
    "=== V2 REJECT CLAIMS ==="
)


semantic_v2_rejects = [
    item
    for item in (
        semantic_v2_judgments
    )
    if item[
        "decision"
    ] == "REJECT"
]


if not semantic_v2_rejects:

    print(
        "  None"
    )


for item in (
    semantic_v2_rejects[:25]
):

    print()
    print(
        "Raw Claim ID:",
        item[
            "raw_claim_id"
        ],
    )

    print(
        "Claim:",
        item[
            "original_claim_text_fa"
        ],
    )

    print(
        "Issues:",
        item[
            "issues"
        ],
    )

    print(
        "Risk signals:",
        item[
            "risk_signals"
        ],
    )

    print(
        "Failed checks:",
        [
            key
            for (
                key,
                value,
            ) in (
                item[
                    "checks"
                ].items()
            )
            if value == "FAIL"
        ],
    )

    print(
        "Reason:",
        item[
            "brief_reason_fa"
        ],
    )


if len(
    semantic_v2_rejects
) > 25:

    print(
        "  ...",
        len(
            semantic_v2_rejects
        ) - 25,
        "additional REJECT judgments not printed",
    )


print()
print(
    "=== V2 ACCEPT SAMPLE ==="
)


semantic_v2_accepts = [
    item
    for item in (
        semantic_v2_judgments
    )
    if item[
        "decision"
    ] == "ACCEPT"
]


for item in (
    semantic_v2_accepts[:10]
):

    print()
    print(
        "Raw Claim ID:",
        item[
            "raw_claim_id"
        ],
    )

    print(
        "Claim:",
        item[
            "original_claim_text_fa"
        ],
    )

    print(
        "Reason:",
        item[
            "brief_reason_fa"
        ],
    )


print()
print(
    "Semantic v2 Artifact:",
    semantic_v2_adjudication_path.relative_to(
        project_root
    ),
)

print(
    "Semantic v2 SHA-256:",
    semantic_v2_adjudication_sha256,
)

print(
    "Write status:",
    semantic_v2_adjudication_write_status,
)


print()
print(
    "Semantic v2 Report:",
    semantic_v2_report_path.relative_to(
        project_root
    ),
)

print(
    "Report write status:",
    semantic_v2_report_write_status,
)


print()
print(
    "Prompt Freeze decision:",
    "PENDING_OUTPUT_REVIEW",
)

print(
    "Full Registry Extraction:",
    "NOT_STARTED",
)


print()
print(
    "CODE CELL 13 GATE:",
    pilot_semantic_v2_status,
)
```

<div style="direction: rtl; white-space: normal; line-height: 1.7;">

<h2>Semantic Fidelity Gate v3 — حفظ Scope، Modality، Perspective و Quantifier منبع</h2>

<p>
Semantic Gate v2 موفق شد Fragmentها، List Headerها، Evidenceهای ناقص و بسیاری از Claimهای Context-dependent را شناسایی کند.
با این حال، بررسی نمونه‌های ACCEPT نشان داد یک Failure Class مهم هنوز باقی مانده است:
Claim ممکن است از همان موضوع Source صحبت کند، اما شدت، دامنه، زاویهٔ بیان یا Quantifier منبع را تغییر دهد.
</p>

<h3>نمونهٔ Failure Class</h3>

<p>
یک عبارت اول‌شخص در پرسشنامه یا مثال شخصی نمی‌تواند بدون Evidence مستقل به گزاره‌ای عمومی دربارهٔ همهٔ سرمایه‌گذاران تبدیل شود.
همچنین Recommendation نباید به Requirement، امکان نباید به قطعیت، و یک Scope محدود مانند «کسب پول بیشتر» نباید به Scope وسیع‌تری مانند «موفقیت مالی» تبدیل شود.
</p>

<h3>هدف این مرحله</h3>

<p>
این مرحله هیچ Claim جدیدی استخراج نمی‌کند و Prompt Extraction v2 را دوباره اجرا نمی‌کند.
همان ۵۷ Raw Claim موجود دوباره adjudicate می‌شوند، اما این بار علاوه بر معیارهای قبلی، وفاداری Claim به نحوهٔ بیان Source نیز بررسی می‌شود.
</p>

<h3>چهار Fidelity Check جدید</h3>

<ul>
<li><strong>Scope Fidelity:</strong> دامنهٔ Claim نباید از دامنهٔ Evidence گسترده‌تر شود.</li>
<li><strong>Modality Fidelity:</strong> امکان، توصیه، الزام، قطعیت، احتمال و رابطهٔ علّی نباید قوی‌تر یا ضعیف‌تر از Source شوند.</li>
<li><strong>Perspective Fidelity:</strong> First-person، Example، Hypothetical، Questionnaire Item یا Attribution نباید بدون Evidence به General Fact تبدیل شوند.</li>
<li><strong>Quantifier Fidelity:</strong> واژه‌هایی مانند «همه»، «معمولاً»، «همیشه»، «فقط»، «اغلب» و مشابه آن فقط زمانی مجازند که Source همان دامنهٔ کمی را پشتیبانی کند.</li>
</ul>

<h3>Source Statement Mode</h3>

<p>
Judge برای هر Claim ابتدا نوع Statement منبع را نیز مشخص می‌کند:
</p>

<ul>
<li><code>GENERAL_DOMAIN_ASSERTION</code></li>
<li><code>RECOMMENDATION_OR_RULE</code></li>
<li><code>DEFINITION</code></li>
<li><code>FIRST_PERSON_SELF_REPORT</code></li>
<li><code>QUESTIONNAIRE_OR_ASSESSMENT_ITEM</code></li>
<li><code>EXAMPLE_OR_HYPOTHETICAL</code></li>
<li><code>SOURCE_COMMENTARY_OR_METADATA</code></li>
<li><code>FRAGMENT_OR_DIALOGUE</code></li>
<li><code>OTHER</code></li>
</ul>

<h3>Repair Contract</h3>

<p>
اگر Claim بیش از Source تعمیم داده شده باشد اما بتوان آن را فقط با Evidence ثبت‌شده به یک گزارهٔ درست و مستقل محدود کرد، <code>REPAIR</code> مجاز است.
</p>

<p>
اگر Repair نیازمند Fact جدیدی از Context پیرامونی باشد، Claim باید <code>REJECT</code> شود.
Surrounding Context فقط برای فهم Source است و Evidence محسوب نمی‌شود.
</p>

<h3>هدف معماری</h3>

<p>
Raw Claim لازم نیست بی‌نقص باشد.
اما Claimی که وارد <code>claims_validated</code> خواهد شد باید هم Grounded و هم Semantically Faithful باشد.
اگر این Gate روی Pilot عملکرد مناسب داشته باشد، Contract زیر Freeze خواهد شد:
</p>

<pre><code>Prompt Window
↓
Raw Claim Extraction v2
↓
Exact Evidence Validation
↓
Semantic Fidelity Gate v3
↓
Validated Claim
</code></pre>

<h3>Artifactهای این مرحله</h3>

<ul>
<li><code>claim_extraction_pilot_semantic_adjudication_v3.jsonl</code></li>
<li><code>claim_extraction_pilot_semantic_report_v3.json</code></li>
<li><code>claim_extraction_pilot_semantic_cache_v3/</code></li>
</ul>

<h3>معیار PASS</h3>

<ul>
<li>هر ۵۷ Claim دقیقاً یک Judgment داشته باشد؛</li>
<li>Scope، Modality، Perspective و Quantifier جداگانه بررسی شوند؛</li>
<li>Questionnaire یا Self-report بدون Evidence مستقل به General Fact تبدیل نشود؛</li>
<li>Repair فقط با Evidenceهای ثبت‌شده انجام شود؛</li>
<li>هیچ Batch Failure حل‌نشده باقی نماند؛</li>
<li>Artifactهای v1 و v2 تغییر نکنند؛</li>
<li>Artifactهای v3 با Policy عدم Overwrite منتشر و از Disk دوباره Validate شوند؛</li>
<li>Full Registry Extraction هنوز در این Cell شروع نشود.</li>
</ul>

</div>

```python
# Code Cell 14

assert (
    pilot_semantic_v2_status
    == "PASS"
), (
    "Code Cell 13 must PASS before "
    "running Semantic Fidelity Gate v3."
)


SEMANTIC_V3_PROMPT_VERSION = (
    "claim_semantic_fidelity_v3"
)

SEMANTIC_V3_BATCH_SIZE = 4
SEMANTIC_V3_MAX_TOKENS = 4500
SEMANTIC_V3_TEMPERATURE = 0
SEMANTIC_V3_MAX_ATTEMPTS = 3
SEMANTIC_V3_CONTEXT_CHARS = 350


SEMANTIC_V3_CHECK_NAMES = {
    "entailment",
    "evidence_completeness",
    "standalone",
    "atomicity",
    "domain_relevance",
    "knowledge_utility",
    "claim_type_fit",
    "scope_fidelity",
    "modality_fidelity",
    "perspective_fidelity",
    "quantifier_fidelity",
}


SEMANTIC_V3_DECISIONS = {
    "ACCEPT",
    "REPAIR",
    "REJECT",
}


SEMANTIC_V3_SOURCE_MODES = {
    "GENERAL_DOMAIN_ASSERTION",
    "RECOMMENDATION_OR_RULE",
    "DEFINITION",
    "FIRST_PERSON_SELF_REPORT",
    "QUESTIONNAIRE_OR_ASSESSMENT_ITEM",
    "EXAMPLE_OR_HYPOTHETICAL",
    "SOURCE_COMMENTARY_OR_METADATA",
    "FRAGMENT_OR_DIALOGUE",
    "OTHER",
}


SEMANTIC_V3_ISSUES = {
    "UNSUPPORTED_OR_OVERSTATED",
    "EVIDENCE_INCOMPLETE",
    "CONTEXT_DEPENDENT",
    "SUBJECT_OR_REFERENT_UNRESOLVED",
    "NOT_ATOMIC",
    "CLAIM_TYPE_MISMATCH",
    "SOURCE_METADATA_OR_COMMENTARY",
    "NON_DOMAIN_KNOWLEDGE",
    "LIST_HEADER_INCOMPLETE",
    "NUMERIC_SUBJECT_MISSING",
    "IMPERATIVE_OR_FRAGMENT",
    "TOO_VAGUE",
    "LOW_KNOWLEDGE_UTILITY",
    "GENERALIZATION_BEYOND_SOURCE",
    "SCOPE_EXPANSION",
    "MODALITY_STRENGTHENED",
    "MODALITY_CHANGED",
    "PERSPECTIVE_SHIFT",
    "QUANTIFIER_SHIFT",
    "QUESTIONNAIRE_OR_SELF_REPORT_AS_GENERAL_FACT",
    "EXAMPLE_OR_HYPOTHETICAL_AS_GENERAL_FACT",
    "OTHER",
}


semantic_v3_adjudication_path = (
    registry_artifact_root
    / "claim_extraction_pilot_semantic_adjudication_v3.jsonl"
)

semantic_v3_report_path = (
    registry_artifact_root
    / "claim_extraction_pilot_semantic_report_v3.json"
)

semantic_v3_cache_root = (
    registry_artifact_root
    / "claim_extraction_pilot_semantic_cache_v3"
)


# ---------------------------------------------------------
# Reload Pilot v2 independently.
# ---------------------------------------------------------

semantic_v3_claims = load_jsonl(
    pilot_v2_raw_path
)

semantic_v3_evidence = load_jsonl(
    pilot_v2_evidence_path
)


assert (
    len(
        semantic_v3_claims
    )
    == 57
)


semantic_v3_claim_by_id = {
    item[
        "raw_claim_id"
    ]: item
    for item in (
        semantic_v3_claims
    )
}


assert (
    len(
        semantic_v3_claim_by_id
    )
    == len(
        semantic_v3_claims
    )
)


semantic_v3_evidence_by_claim = defaultdict(
    list
)


for evidence in (
    semantic_v3_evidence
):

    assert (
        evidence[
            "raw_claim_id"
        ]
        in semantic_v3_claim_by_id
    )

    semantic_v3_evidence_by_claim[
        evidence[
            "raw_claim_id"
        ]
    ].append(
        evidence
    )


for raw_claim_id in (
    semantic_v3_claim_by_id
):

    assert (
        semantic_v3_evidence_by_claim[
            raw_claim_id
        ]
    )


# ---------------------------------------------------------
# Deterministic fidelity-risk signals.
# These are warnings for the judge, not automatic decisions.
# ---------------------------------------------------------

def normalize_fidelity_text(
    text,
):
    return (
        " ".join(
            str(
                text
            ).strip().split()
        )
    )


def detect_fidelity_risk_signals(
    claim_text,
    evidence_quotes,
    surrounding_contexts,
):
    claim = normalize_fidelity_text(
        claim_text
    )

    evidence_text = " ".join(
        normalize_fidelity_text(
            item
        )
        for item in evidence_quotes
    )

    context_text = " ".join(
        normalize_fidelity_text(
            item
        )
        for item in surrounding_contexts
    )


    signals = set(
        detect_claim_risk_signals(
            claim
        )
    )


    first_person_markers = [
        " من ",
        " می‌خواهم ",
        " ترجیح می‌دهم ",
        " انتخاب می‌کنم ",
        " دوست دارم ",
        " فکر می‌کنم ",
        " به نظر من ",
    ]


    padded_evidence = (
        f" {evidence_text} "
    )


    if any(
        marker
        in padded_evidence
        for marker in (
            first_person_markers
        )
    ):
        signals.add(
            "SOURCE_FIRST_PERSON"
        )


    questionnaire_markers = [
        "پرسشنامه",
        "سؤال",
        "سوال",
        "گزینه",
        "امتیاز",
    ]


    if (
        any(
            marker
            in context_text
            for marker in (
                questionnaire_markers
            )
        )
        or (
            re.match(
                r"^\s*[\d۰-۹]+[\.\-\)]",
                evidence_text,
            )
            is not None
            and "می‌" in evidence_text
        )
    ):
        signals.add(
            "POSSIBLE_QUESTIONNAIRE_OR_SELF_REPORT"
        )


    strong_modality_terms = [
        "ضروری است",
        "الزامی است",
        "حتماً",
        "باید",
        "قطعا",
        "قطعاً",
        "همیشه",
    ]


    if any(
        term in claim
        for term in (
            strong_modality_terms
        )
    ) and not any(
        term in evidence_text
        for term in (
            strong_modality_terms
        )
    ):
        signals.add(
            "POSSIBLE_MODALITY_STRENGTHENING"
        )


    quantifier_terms = [
        "همه",
        "همیشه",
        "هرگز",
        "فقط",
        "معمولاً",
        "معمولا",
        "اغلب",
        "عمدتاً",
        "عمدتا",
        "تمام",
    ]


    for term in quantifier_terms:

        if (
            term in claim
            and term not in evidence_text
        ):
            signals.add(
                "POSSIBLE_QUANTIFIER_SHIFT"
            )

            break


    generalized_subject_terms = [
        "سرمایه‌گذاران",
        "افراد",
        "مردم",
        "کودکان",
        "بانک‌ها",
        "خانوارها",
    ]


    if (
        "SOURCE_FIRST_PERSON"
        in signals
        and any(
            term in claim
            for term in (
                generalized_subject_terms
            )
        )
    ):
        signals.add(
            "POSSIBLE_FIRST_PERSON_GENERALIZATION"
        )


    return sorted(
        signals
    )


# ---------------------------------------------------------
# Evidence + judge-only surrounding context.
# ---------------------------------------------------------

def build_semantic_v3_evidence_item(
    evidence,
):
    document_id = evidence[
        "document_id"
    ]

    source_text = (
        accepted_records_by_id[
            document_id
        ][
            canonical_text_field
        ]
    )

    evidence_start = (
        evidence[
            "locator"
        ][
            "text_char_start"
        ]
    )

    evidence_end = (
        evidence[
            "locator"
        ][
            "text_char_end"
        ]
    )

    quote = evidence[
        "evidence_quote_exact"
    ]


    assert (
        source_text[
            evidence_start:
            evidence_end
        ]
        == quote
    )


    context_start = max(
        0,
        evidence_start
        - SEMANTIC_V3_CONTEXT_CHARS,
    )

    context_end = min(
        len(
            source_text
        ),
        evidence_end
        + SEMANTIC_V3_CONTEXT_CHARS,
    )


    return {
        "evidence_id": (
            evidence[
                "evidence_id"
            ]
        ),
        "document_id": (
            document_id
        ),
        "source_type": (
            evidence[
                "source_type"
            ]
        ),
        "quote_exact": (
            quote
        ),
        "judge_only_surrounding_context": (
            source_text[
                context_start:
                context_end
            ]
        ),
    }


semantic_v3_items = []


for claim in sorted(
    semantic_v3_claims,
    key=lambda item: (
        item[
            "raw_claim_id"
        ]
    ),
):

    raw_claim_id = claim[
        "raw_claim_id"
    ]


    evidence_items = [
        build_semantic_v3_evidence_item(
            evidence
        )
        for evidence in (
            semantic_v3_evidence_by_claim[
                raw_claim_id
            ]
        )
    ]


    evidence_quotes = [
        item[
            "quote_exact"
        ]
        for item in (
            evidence_items
        )
    ]


    surrounding_contexts = [
        item[
            "judge_only_surrounding_context"
        ]
        for item in (
            evidence_items
        )
    ]


    semantic_v3_items.append({
        "raw_claim_id": (
            raw_claim_id
        ),
        "claim_text_fa": (
            claim[
                "claim_text_fa"
            ]
        ),
        "claim_type": (
            claim[
                "claim_type"
            ]
        ),
        "authority_class": (
            claim[
                "authority_class"
            ]
        ),
        "claim_discovery_role": (
            claim[
                "claim_discovery_role"
            ]
        ),
        "fidelity_risk_signals": (
            detect_fidelity_risk_signals(
                claim[
                    "claim_text_fa"
                ],
                evidence_quotes,
                surrounding_contexts,
            )
        ),
        "evidence": (
            evidence_items
        ),
    })


assert (
    len(
        semantic_v3_items
    )
    == 57
)


semantic_v3_batches = [
    semantic_v3_items[
        start:
        start
        + SEMANTIC_V3_BATCH_SIZE
    ]
    for start in range(
        0,
        len(
            semantic_v3_items
        ),
        SEMANTIC_V3_BATCH_SIZE,
    )
]


def semantic_v3_batch_id(
    batch_items,
):
    material = stable_json({
        "prompt_version": (
            SEMANTIC_V3_PROMPT_VERSION
        ),
        "model": (
            pilot_chat_model
        ),
        "claim_ids": [
            item[
                "raw_claim_id"
            ]
            for item in (
                batch_items
            )
        ],
    })


    digest = hashlib.sha256(
        material.encode(
            "utf-8"
        )
    ).hexdigest()[:24]


    return (
        f"semantic-v3-batch:{digest}"
    )


# ---------------------------------------------------------
# Semantic Fidelity prompt.
# ---------------------------------------------------------

def build_semantic_v3_prompt(
    batch_items,
    retry_feedback=None,
):
    retry_block = ""


    if retry_feedback:

        retry_block = f"""

Your previous response failed deterministic validation.

Validator error:
{retry_feedback}

Return the COMPLETE corrected JSON for every raw_claim_id.
"""


    system_message = """
You are the final semantic-fidelity adjudicator for a Persian financial knowledge registry.

You are NOT extracting new claims.
You must decide whether each existing Claim faithfully represents its REGISTERED Evidence.

Use only:
1. the Claim,
2. its exact registered Evidence,
3. limited surrounding context for understanding source discourse.

Surrounding context is NOT Evidence.
It cannot supply facts to an ACCEPT or REPAIR decision.

A Claim can be factually related to the source and still be invalid if it changes:
- scope,
- certainty or modality,
- speaker perspective,
- quantifiers,
- source statement mode.

SOURCE-MODE RULES:

FIRST_PERSON_SELF_REPORT:
A statement such as "I prefer..." describes the respondent or questionnaire statement.
It cannot become "investors usually prefer..." without explicit evidence for that generalization.

QUESTIONNAIRE_OR_ASSESSMENT_ITEM:
A questionnaire item is not automatically a general fact about people or investors.
Reject generalized population Claims derived only from an assessment item.

EXAMPLE_OR_HYPOTHETICAL:
An example cannot become a universal rule unless the Evidence explicitly states the rule.

RECOMMENDATION_OR_RULE:
Do not strengthen "consider", "can", or an ordinary recommendation into "must", "necessary", "always", or a legal/absolute requirement.

SCOPE RULE:
Do not broaden a narrow concept.
For example, a statement about "earning more money" does not automatically support a broader statement about "financial success".

QUANTIFIER RULE:
Do not introduce "all", "always", "usually", "only", "most", "never" or similar quantifiers unless Evidence supports them.

CAUSALITY RULE:
Do not turn association, advice, sequence, or possibility into a causal law.

REPAIR:
Repair is allowed only when all facts in the repaired Claim are already present in REGISTERED Evidence.
You may narrow wording, preserve original perspective, remove overstatement, or correct claim_type.
You may not import a missing subject or fact only from surrounding context.

REJECT:
Reject when a faithful standalone domain Claim cannot be formed from registered Evidence alone.

Return valid JSON only.
No markdown.
No text outside JSON.
""".strip()


    user_message = f"""
Adjudicate every Claim.

Allowed decisions:
{json.dumps(sorted(SEMANTIC_V3_DECISIONS), ensure_ascii=False)}

Allowed source_statement_mode values:
{json.dumps(sorted(SEMANTIC_V3_SOURCE_MODES), ensure_ascii=False)}

Required checks:
{json.dumps(sorted(SEMANTIC_V3_CHECK_NAMES), ensure_ascii=False)}

Each check value:
PASS or FAIL

Allowed issues:
{json.dumps(sorted(SEMANTIC_V3_ISSUES), ensure_ascii=False)}

Allowed Claim Types:
{json.dumps(sorted(ALLOWED_CLAIM_TYPES), ensure_ascii=False)}

Required JSON schema:

{{
  "judgments": [
    {{
      "raw_claim_id": "raw-claim:...",
      "source_statement_mode": "ONE_ALLOWED_MODE",
      "decision": "ACCEPT | REPAIR | REJECT",
      "checks": {{
        "entailment": "PASS | FAIL",
        "evidence_completeness": "PASS | FAIL",
        "standalone": "PASS | FAIL",
        "atomicity": "PASS | FAIL",
        "domain_relevance": "PASS | FAIL",
        "knowledge_utility": "PASS | FAIL",
        "claim_type_fit": "PASS | FAIL",
        "scope_fidelity": "PASS | FAIL",
        "modality_fidelity": "PASS | FAIL",
        "perspective_fidelity": "PASS | FAIL",
        "quantifier_fidelity": "PASS | FAIL"
      }},
      "issues": [],
      "repaired_claim_text_fa": null,
      "recommended_claim_type": null,
      "repair_basis_evidence_ids": [],
      "brief_reason_fa": "دلیل کوتاه و مشخص"
    }}
  ]
}}

Decision rules:

ACCEPT:
- every check must PASS,
- repaired_claim_text_fa = null,
- recommended_claim_type = original claim_type,
- repair_basis_evidence_ids = [].

REPAIR:
- entailment = PASS,
- evidence_completeness = PASS,
- domain_relevance = PASS,
- knowledge_utility = PASS,
- repaired_claim_text_fa must be non-empty,
- recommended_claim_type must be an allowed type,
- repair_basis_evidence_ids must reference existing registered Evidence only.

REJECT:
- repaired_claim_text_fa = null,
- recommended_claim_type = null,
- repair_basis_evidence_ids = [].

Important:
A questionnaire/self-report/example that was generalized beyond its Evidence must NOT be ACCEPTED.
A stronger modality than Evidence must NOT be ACCEPTED.
A broader scope than Evidence must NOT be ACCEPTED.
A new quantifier absent from Evidence must NOT be ACCEPTED.

Items:

{json.dumps(batch_items, ensure_ascii=False, indent=2)}

{retry_block}
""".strip()


    return (
        system_message,
        user_message,
    )


def call_semantic_v3_judge(
    system_message,
    user_message,
):
    response = (
        pilot_client
        .chat
        .completions
        .create(
            model=(
                pilot_chat_model
            ),
            messages=[
                {
                    "role": "system",
                    "content": (
                        system_message
                    ),
                },
                {
                    "role": "user",
                    "content": (
                        user_message
                    ),
                },
            ],
            temperature=(
                SEMANTIC_V3_TEMPERATURE
            ),
            max_tokens=(
                SEMANTIC_V3_MAX_TOKENS
            ),
        )
    )


    content = (
        response
        .choices[0]
        .message
        .content
    )


    if not isinstance(
        content,
        str,
    ):
        raise ValueError(
            "Semantic v3 response "
            "content is not a string."
        )


    return content


# ---------------------------------------------------------
# Deterministic validation.
# ---------------------------------------------------------

def validate_semantic_v3_payload(
    batch_items,
    payload,
):
    if not isinstance(
        payload,
        dict,
    ):
        raise ValueError(
            "Semantic v3 payload "
            "must be an object."
        )


    judgments = payload.get(
        "judgments"
    )


    if not isinstance(
        judgments,
        list,
    ):
        raise ValueError(
            "'judgments' must be a list."
        )


    expected_ids = {
        item[
            "raw_claim_id"
        ]
        for item in (
            batch_items
        )
    }


    original_by_id = {
        item[
            "raw_claim_id"
        ]: item
        for item in (
            batch_items
        )
    }


    observed_ids = [
        item.get(
            "raw_claim_id"
        )
        if isinstance(
            item,
            dict,
        )
        else None
        for item in (
            judgments
        )
    ]


    if (
        len(
            observed_ids
        )
        != len(
            set(
                observed_ids
            )
        )
    ):
        raise ValueError(
            "Duplicate raw_claim_id "
            "in Semantic v3 response."
        )


    if (
        set(
            observed_ids
        )
        != expected_ids
    ):
        raise ValueError(
            "Semantic v3 Claim IDs "
            "do not exactly match batch."
        )


    validated = []


    for judgment in judgments:

        if not isinstance(
            judgment,
            dict,
        ):
            raise ValueError(
                "Judgment must be an object."
            )


        raw_claim_id = (
            judgment[
                "raw_claim_id"
            ]
        )

        original = (
            original_by_id[
                raw_claim_id
            ]
        )


        source_mode = str(
            judgment.get(
                "source_statement_mode",
                "",
            )
        ).strip()


        if (
            source_mode
            not in SEMANTIC_V3_SOURCE_MODES
        ):
            raise ValueError(
                "Invalid source_statement_mode: "
                f"{source_mode!r}"
            )


        decision = str(
            judgment.get(
                "decision",
                "",
            )
        ).strip()


        if (
            decision
            not in SEMANTIC_V3_DECISIONS
        ):
            raise ValueError(
                "Invalid decision: "
                f"{decision!r}"
            )


        checks = judgment.get(
            "checks"
        )


        if not isinstance(
            checks,
            dict,
        ):
            raise ValueError(
                "Checks must be an object."
            )


        if (
            set(
                checks
            )
            != SEMANTIC_V3_CHECK_NAMES
        ):
            raise ValueError(
                "Semantic v3 check keys "
                "do not match Contract."
            )


        for (
            check_name,
            check_value,
        ) in checks.items():

            if (
                check_value
                not in {
                    "PASS",
                    "FAIL",
                }
            ):
                raise ValueError(
                    f"Invalid {check_name}: "
                    f"{check_value!r}"
                )


        issues = judgment.get(
            "issues",
            []
        )


        if not isinstance(
            issues,
            list,
        ):
            raise ValueError(
                "Issues must be a list."
            )


        if (
            len(
                issues
            )
            != len(
                set(
                    issues
                )
            )
        ):
            raise ValueError(
                "Duplicate Semantic v3 issue."
            )


        unknown_issues = (
            set(
                issues
            )
            - SEMANTIC_V3_ISSUES
        )


        if unknown_issues:
            raise ValueError(
                "Unknown Semantic v3 issues: "
                f"{sorted(unknown_issues)}"
            )


        repaired_claim = judgment.get(
            "repaired_claim_text_fa"
        )

        recommended_type = judgment.get(
            "recommended_claim_type"
        )

        repair_basis = judgment.get(
            "repair_basis_evidence_ids",
            [],
        )


        if not isinstance(
            repair_basis,
            list,
        ):
            raise ValueError(
                "repair_basis_evidence_ids "
                "must be a list."
            )


        allowed_evidence_ids = {
            item[
                "evidence_id"
            ]
            for item in (
                original[
                    "evidence"
                ]
            )
        }


        if not set(
            repair_basis
        ).issubset(
            allowed_evidence_ids
        ):
            raise ValueError(
                "Repair references Evidence "
                "outside this Claim."
            )


        reason = str(
            judgment.get(
                "brief_reason_fa",
                "",
            )
        ).strip()


        if not reason:
            raise ValueError(
                "brief_reason_fa is empty."
            )


        if decision == "ACCEPT":

            if any(
                value != "PASS"
                for value in (
                    checks.values()
                )
            ):
                raise ValueError(
                    "ACCEPT requires "
                    "all checks PASS."
                )


            if repaired_claim is not None:
                raise ValueError(
                    "ACCEPT may not include "
                    "a repaired Claim."
                )


            if (
                recommended_type
                != original[
                    "claim_type"
                ]
            ):
                raise ValueError(
                    "ACCEPT must preserve "
                    "original Claim Type."
                )


            if repair_basis:
                raise ValueError(
                    "ACCEPT must have "
                    "empty repair basis."
                )


        elif decision == "REPAIR":

            for required_pass in [
                "entailment",
                "evidence_completeness",
                "domain_relevance",
                "knowledge_utility",
            ]:

                if (
                    checks[
                        required_pass
                    ]
                    != "PASS"
                ):
                    raise ValueError(
                        "REPAIR requires "
                        f"{required_pass}=PASS."
                    )


            if (
                not isinstance(
                    repaired_claim,
                    str,
                )
                or not repaired_claim.strip()
            ):
                raise ValueError(
                    "REPAIR requires "
                    "a repaired Claim."
                )


            repaired_claim = (
                repaired_claim.strip()
            )


            if (
                recommended_type
                not in ALLOWED_CLAIM_TYPES
            ):
                raise ValueError(
                    "REPAIR has invalid "
                    "recommended Claim Type."
                )


            if not repair_basis:
                raise ValueError(
                    "REPAIR requires "
                    "existing Evidence IDs."
                )


        else:

            if repaired_claim is not None:
                raise ValueError(
                    "REJECT must not include "
                    "a repaired Claim."
                )


            if recommended_type is not None:
                raise ValueError(
                    "REJECT must use "
                    "recommended_claim_type=null."
                )


            if repair_basis:
                raise ValueError(
                    "REJECT must have "
                    "empty repair basis."
                )


        issue_check_contracts = {
            "EVIDENCE_INCOMPLETE": (
                "evidence_completeness"
            ),
            "CLAIM_TYPE_MISMATCH": (
                "claim_type_fit"
            ),
            "NON_DOMAIN_KNOWLEDGE": (
                "domain_relevance"
            ),
            "LOW_KNOWLEDGE_UTILITY": (
                "knowledge_utility"
            ),
            "SCOPE_EXPANSION": (
                "scope_fidelity"
            ),
            "MODALITY_STRENGTHENED": (
                "modality_fidelity"
            ),
            "MODALITY_CHANGED": (
                "modality_fidelity"
            ),
            "PERSPECTIVE_SHIFT": (
                "perspective_fidelity"
            ),
            "QUANTIFIER_SHIFT": (
                "quantifier_fidelity"
            ),
        }


        for (
            issue,
            required_failed_check,
        ) in (
            issue_check_contracts.items()
        ):

            if (
                issue in issues
                and checks[
                    required_failed_check
                ]
                != "FAIL"
            ):
                raise ValueError(
                    f"{issue} requires "
                    f"{required_failed_check}=FAIL."
                )


        if (
            "UNSUPPORTED_OR_OVERSTATED"
            in issues
            and checks[
                "entailment"
            ]
            != "FAIL"
        ):
            raise ValueError(
                "UNSUPPORTED_OR_OVERSTATED "
                "requires entailment=FAIL."
            )


        validated.append({
            "raw_claim_id": (
                raw_claim_id
            ),
            "source_statement_mode": (
                source_mode
            ),
            "decision": (
                decision
            ),
            "checks": {
                key: checks[
                    key
                ]
                for key in sorted(
                    SEMANTIC_V3_CHECK_NAMES
                )
            },
            "issues": sorted(
                issues
            ),
            "fidelity_risk_signals": (
                original[
                    "fidelity_risk_signals"
                ]
            ),
            "repaired_claim_text_fa": (
                repaired_claim
            ),
            "recommended_claim_type": (
                recommended_type
            ),
            "repair_basis_evidence_ids": (
                sorted(
                    repair_basis
                )
            ),
            "brief_reason_fa": (
                reason
            ),
            "original_claim_text_fa": (
                original[
                    "claim_text_fa"
                ]
            ),
            "original_claim_type": (
                original[
                    "claim_type"
                ]
            ),
            "semantic_model": (
                pilot_chat_model
            ),
            "semantic_prompt_version": (
                SEMANTIC_V3_PROMPT_VERSION
            ),
        })


    return sorted(
        validated,
        key=lambda item: (
            item[
                "raw_claim_id"
            ]
        ),
    )


# ---------------------------------------------------------
# Execute with validated cache.
# ---------------------------------------------------------

semantic_v3_cache_root.mkdir(
    parents=True,
    exist_ok=True,
)


semantic_v3_judgments = []
semantic_v3_failures = []

semantic_v3_api_calls_this_run = 0
semantic_v3_cache_hits_this_run = 0


print(
    "=== NB12i SEMANTIC FIDELITY GATE V3 ==="
)

print()
print(
    "Pilot v2 Claims:",
    len(
        semantic_v3_items
    ),
)

print(
    "Semantic v3 batches:",
    len(
        semantic_v3_batches
    ),
)

print()


for batch_number, batch_items in enumerate(
    semantic_v3_batches,
    1,
):

    batch_id = (
        semantic_v3_batch_id(
            batch_items
        )
    )


    cache_path = (
        semantic_v3_cache_root
        / (
            batch_id.replace(
                ":",
                "_",
            )
            + ".json"
        )
    )


    print(
        f"[{batch_number:02d}/{len(semantic_v3_batches):02d}]",
        batch_id,
        "| claims:",
        len(
            batch_items
        ),
    )


    if cache_path.exists():

        cached = load_json(
            cache_path
        )


        assert (
            cached[
                "batch_id"
            ]
            == batch_id
        )


        assert (
            cached[
                "semantic_prompt_version"
            ]
            == SEMANTIC_V3_PROMPT_VERSION
        )


        assert (
            cached[
                "model"
            ]
            == pilot_chat_model
        )


        cached_judgments = (
            cached[
                "judgments"
            ]
        )


        assert (
            {
                item[
                    "raw_claim_id"
                ]
                for item in (
                    cached_judgments
                )
            }
            == {
                item[
                    "raw_claim_id"
                ]
                for item in (
                    batch_items
                )
            }
        )


        semantic_v3_judgments.extend(
            cached_judgments
        )


        semantic_v3_cache_hits_this_run += 1


        print(
            "   cache: HIT"
        )


        continue


    validated_judgments = None
    last_error = None


    for attempt in range(
        1,
        SEMANTIC_V3_MAX_ATTEMPTS + 1,
    ):

        (
            system_message,
            user_message,
        ) = build_semantic_v3_prompt(
            batch_items,
            retry_feedback=(
                last_error
            ),
        )


        try:

            content = (
                call_semantic_v3_judge(
                    system_message,
                    user_message,
                )
            )


            semantic_v3_api_calls_this_run += 1


            payload = parse_model_json(
                content
            )


            validated_judgments = (
                validate_semantic_v3_payload(
                    batch_items,
                    payload,
                )
            )


            break


        except Exception as exc:

            last_error = (
                f"{type(exc).__name__}: "
                f"{exc}"
            )


            print(
                f"   attempt {attempt}:",
                "INVALID",
                "|",
                last_error,
            )


            if (
                attempt
                < SEMANTIC_V3_MAX_ATTEMPTS
            ):
                time.sleep(
                    min(
                        2.0,
                        float(
                            attempt
                        ),
                    )
                )


    if (
        validated_judgments
        is None
    ):

        semantic_v3_failures.append({
            "batch_id": (
                batch_id
            ),
            "raw_claim_ids": [
                item[
                    "raw_claim_id"
                ]
                for item in (
                    batch_items
                )
            ],
            "last_error": (
                last_error
            ),
        })


        print(
            "   result: FAIL"
        )


        continue


    cache_payload = {
        "batch_id": (
            batch_id
        ),
        "semantic_prompt_version": (
            SEMANTIC_V3_PROMPT_VERSION
        ),
        "model": (
            pilot_chat_model
        ),
        "judgments": (
            validated_judgments
        ),
    }


    cache_text = (
        json.dumps(
            cache_payload,
            ensure_ascii=False,
            indent=2,
            sort_keys=True,
        )
        + "\n"
    )


    with cache_path.open(
        "x",
        encoding="utf-8",
    ) as file:
        file.write(
            cache_text
        )


    semantic_v3_judgments.extend(
        validated_judgments
    )


    print(
        "   result: VALID"
    )


print()


assert not semantic_v3_failures, (
    "Semantic v3 has unresolved failures:\n"
    + json.dumps(
        semantic_v3_failures,
        ensure_ascii=False,
        indent=2,
    )
)


semantic_v3_judgments = sorted(
    semantic_v3_judgments,
    key=lambda item: (
        item[
            "raw_claim_id"
        ]
    ),
)


semantic_v3_ids = [
    item[
        "raw_claim_id"
    ]
    for item in (
        semantic_v3_judgments
    )
]


assert (
    len(
        semantic_v3_ids
    )
    == 57
)


assert (
    len(
        semantic_v3_ids
    )
    == len(
        set(
            semantic_v3_ids
        )
    )
)


assert (
    set(
        semantic_v3_ids
    )
    == set(
        semantic_v3_claim_by_id
    )
)


# ---------------------------------------------------------
# Aggregate.
# ---------------------------------------------------------

semantic_v3_decision_counts = Counter(
    item[
        "decision"
    ]
    for item in (
        semantic_v3_judgments
    )
)


semantic_v3_issue_counts = Counter()

semantic_v3_failed_check_counts = Counter()

semantic_v3_source_mode_counts = Counter()


for item in (
    semantic_v3_judgments
):

    semantic_v3_source_mode_counts[
        item[
            "source_statement_mode"
        ]
    ] += 1


    for issue in (
        item[
            "issues"
        ]
    ):
        semantic_v3_issue_counts[
            issue
        ] += 1


    for (
        check_name,
        check_value,
    ) in (
        item[
            "checks"
        ].items()
    ):

        if (
            check_value
            == "FAIL"
        ):
            semantic_v3_failed_check_counts[
                check_name
            ] += 1


# ---------------------------------------------------------
# Publish.
# ---------------------------------------------------------

semantic_v3_text = "".join(
    json.dumps(
        item,
        ensure_ascii=False,
        sort_keys=True,
    )
    + "\n"
    for item in (
        semantic_v3_judgments
    )
)


semantic_v3_sha256 = hashlib.sha256(
    semantic_v3_text.encode(
        "utf-8"
    )
).hexdigest()


semantic_v3_report = {
    "report_version": (
        "claim_extraction_pilot_semantic_report_v3"
    ),

    "source_pilot": (
        pilot_v2_report_path.name
    ),

    "source_raw_claims_sha256": (
        file_sha256(
            pilot_v2_raw_path
        )
    ),

    "semantic_prompt_version": (
        SEMANTIC_V3_PROMPT_VERSION
    ),

    "semantic_model": (
        pilot_chat_model
    ),

    "claims_adjudicated": (
        len(
            semantic_v3_judgments
        )
    ),

    "decision_counts": dict(
        sorted(
            semantic_v3_decision_counts.items()
        )
    ),

    "failed_checks": dict(
        sorted(
            semantic_v3_failed_check_counts.items()
        )
    ),

    "issue_counts": dict(
        sorted(
            semantic_v3_issue_counts.items()
        )
    ),

    "source_statement_modes": dict(
        sorted(
            semantic_v3_source_mode_counts.items()
        )
    ),

    "semantic_v2_decisions": dict(
        sorted(
            semantic_v2_decision_counts.items()
        )
    ),

    "unresolved_batch_failures": 0,

    "full_registry_readiness": (
        "AWAITING_FINAL_PILOT_REVIEW"
    ),

    "full_registry_extraction_started": False,

    "artifact": {
        "adjudication": (
            semantic_v3_adjudication_path.name
        ),
        "sha256": (
            semantic_v3_sha256
        ),
    },
}


semantic_v3_report_text = (
    json.dumps(
        semantic_v3_report,
        ensure_ascii=False,
        indent=2,
        sort_keys=True,
    )
    + "\n"
)


semantic_v3_write_status = (
    publish_semantic_no_overwrite(
        semantic_v3_adjudication_path,
        semantic_v3_text,
    )
)


semantic_v3_report_write_status = (
    publish_semantic_no_overwrite(
        semantic_v3_report_path,
        semantic_v3_report_text,
    )
)


assert (
    load_jsonl(
        semantic_v3_adjudication_path
    )
    == semantic_v3_judgments
)


assert (
    load_json(
        semantic_v3_report_path
    )
    == semantic_v3_report
)


assert (
    file_sha256(
        semantic_v3_adjudication_path
    )
    == semantic_v3_sha256
)


pilot_semantic_v3_status = (
    "PASS"
)


# ---------------------------------------------------------
# Readable output.
# ---------------------------------------------------------

print(
    "=== NB12i SEMANTIC FIDELITY RESULT V3 ==="
)

print()
print(
    "Claims adjudicated:",
    len(
        semantic_v3_judgments
    ),
)

print(
    "API calls this run:",
    semantic_v3_api_calls_this_run,
)

print(
    "Validated cache hits:",
    semantic_v3_cache_hits_this_run,
)


print()
print(
    "Semantic v2 decisions:",
    dict(
        sorted(
            semantic_v2_decision_counts.items()
        )
    ),
)

print(
    "Semantic v3 decisions:",
    dict(
        sorted(
            semantic_v3_decision_counts.items()
        )
    ),
)


print()
print(
    "Failed checks:",
    dict(
        sorted(
            semantic_v3_failed_check_counts.items()
        )
    ),
)


print()
print(
    "Issue counts:",
    dict(
        sorted(
            semantic_v3_issue_counts.items()
        )
    ),
)


print()
print(
    "Source statement modes:",
    dict(
        sorted(
            semantic_v3_source_mode_counts.items()
        )
    ),
)


print()
print(
    "=== V3 REPAIR CLAIMS ==="
)


v3_repairs = [
    item
    for item in (
        semantic_v3_judgments
    )
    if item[
        "decision"
    ] == "REPAIR"
]


if not v3_repairs:
    print(
        "  None"
    )


for item in (
    v3_repairs[:20]
):

    print()
    print(
        "Raw Claim ID:",
        item[
            "raw_claim_id"
        ],
    )

    print(
        "Source mode:",
        item[
            "source_statement_mode"
        ],
    )

    print(
        "Original:",
        item[
            "original_claim_text_fa"
        ],
    )

    print(
        "Repair:",
        item[
            "repaired_claim_text_fa"
        ],
    )

    print(
        "Issues:",
        item[
            "issues"
        ],
    )

    print(
        "Reason:",
        item[
            "brief_reason_fa"
        ],
    )


print()
print(
    "=== V3 REJECT CLAIMS ==="
)


v3_rejects = [
    item
    for item in (
        semantic_v3_judgments
    )
    if item[
        "decision"
    ] == "REJECT"
]


if not v3_rejects:
    print(
        "  None"
    )


for item in (
    v3_rejects[:25]
):

    print()
    print(
        "Raw Claim ID:",
        item[
            "raw_claim_id"
        ],
    )

    print(
        "Source mode:",
        item[
            "source_statement_mode"
        ],
    )

    print(
        "Claim:",
        item[
            "original_claim_text_fa"
        ],
    )

    print(
        "Issues:",
        item[
            "issues"
        ],
    )

    print(
        "Failed checks:",
        [
            key
            for (
                key,
                value,
            ) in (
                item[
                    "checks"
                ].items()
            )
            if value == "FAIL"
        ],
    )

    print(
        "Reason:",
        item[
            "brief_reason_fa"
        ],
    )


print()
print(
    "=== V3 ACCEPT SAMPLE ==="
)


v3_accepts = [
    item
    for item in (
        semantic_v3_judgments
    )
    if item[
        "decision"
    ] == "ACCEPT"
]


for item in (
    v3_accepts[:12]
):

    print()
    print(
        "Raw Claim ID:",
        item[
            "raw_claim_id"
        ],
    )

    print(
        "Source mode:",
        item[
            "source_statement_mode"
        ],
    )

    print(
        "Claim:",
        item[
            "original_claim_text_fa"
        ],
    )

    print(
        "Reason:",
        item[
            "brief_reason_fa"
        ],
    )


print()
print(
    "Semantic v3 Artifact:",
    semantic_v3_adjudication_path.relative_to(
        project_root
    ),
)

print(
    "Semantic v3 SHA-256:",
    semantic_v3_sha256,
)

print(
    "Write status:",
    semantic_v3_write_status,
)


print()
print(
    "Semantic v3 Report:",
    semantic_v3_report_path.relative_to(
        project_root
    ),
)

print(
    "Report write status:",
    semantic_v3_report_write_status,
)


print()
print(
    "Full Registry decision:",
    "PENDING_FINAL_PILOT_REVIEW",
)

print(
    "Full Registry Extraction:",
    "NOT_STARTED",
)


print()
print(
    "CODE CELL 14 GATE:",
    pilot_semantic_v3_status,
)
```

<div style="direction: rtl; white-space: normal; line-height: 1.7;">

<h2>Freeze قرارداد نهایی Claim Extraction و Validation</h2>

<p>
سه مرحلهٔ Pilot نشان دادند که جداسازی Raw Extraction از Validated Registry ضروری است.
Raw Claim می‌تواند Candidate مفیدی باشد حتی اگر هنوز برای استفادهٔ مستقیم در Golden Dataset، Book یا Knowledge Graph مناسب نباشد.
</p>

<p>
بنابراین Architecture نهایی Claim Registry دو سطح مستقل خواهد داشت:
</p>

<pre><code>Prompt Window
↓
Raw Claim Extraction v2
↓
Exact Evidence Validation
↓
Semantic Fidelity Gate v3
↓
Deterministic Critical-Risk Guard
↓
Validated Claim
</code></pre>

<h3>اصل حفظ Recall</h3>

<p>
Claimهایی که Semantic Gate آن‌ها را Reject یا Hold می‌کند حذف فیزیکی نمی‌شوند.
آن‌ها همچنان در <code>claims_raw_v1.jsonl</code> باقی خواهند ماند تا Provenance و امکان بازبینی بعدی حفظ شود.
</p>

<h3>اصل Precision در Validated Registry</h3>

<p>
ورود خودکار به <code>claims_validated_v1.jsonl</code> فقط زمانی مجاز است که:
</p>

<ul>
<li>Exact Evidence Validation PASS باشد؛</li>
<li>Semantic Fidelity Gate v3 تصمیم <code>ACCEPT</code> داده باشد؛</li>
<li>هیچ Critical Structural Risk Signal فعال نباشد.</li>
</ul>

<h3>Critical Risk Signalهای نهایی</h3>

<p>
Signalهای زیر به اندازهٔ کافی ساختاری و High-Precision هستند که حتی در صورت ACCEPT شدن توسط LLM، Claim را از Auto-validation خارج کنند:
</p>

<ul>
<li><code>FRAGMENT_OR_DISCOURSE_START</code></li>
<li><code>SHORT_NUMERIC_CLAIM</code></li>
<li><code>POSSIBLE_LIST_HEADER</code></li>
<li><code>POSSIBLE_SOURCE_COMMENTARY</code></li>
<li><code>SOURCE_FIRST_PERSON</code></li>
<li><code>POSSIBLE_QUESTIONNAIRE_OR_SELF_REPORT</code></li>
<li><code>POSSIBLE_FIRST_PERSON_GENERALIZATION</code></li>
</ul>

<p>
این Signalها باعث حذف Raw Claim نمی‌شوند؛ Claim به وضعیت Hold منتقل می‌شود.
</p>

<h3>سیاست REPAIR</h3>

<p>
Repairهای پیشنهادشده توسط LLM به‌صورت خودکار وارد Validated Registry نمی‌شوند.
آن‌ها در وضعیت <code>HOLD_REPAIR_REQUIRED</code> باقی می‌مانند.
در نتیجه هیچ Fact بازنویسی‌شده بدون مرحلهٔ مستقل Validation به Registry نهایی وارد نمی‌شود.
</p>

<h3>Dispositionهای نهایی</h3>

<ul>
<li><code>VALIDATED</code> — مناسب ورود خودکار به Validated Registry؛</li>
<li><code>HOLD_CRITICAL_RISK</code> — Semantic ACCEPT شده اما Risk ساختاری دارد؛</li>
<li><code>HOLD_REPAIR_REQUIRED</code> — Semantic Gate پیشنهاد Repair داده است؛</li>
<li><code>REJECT_SEMANTIC</code> — برای Validated Registry مناسب نیست، اما در Raw Registry حفظ می‌شود.</li>
</ul>

<h3>Contract Full Registry</h3>

<p>
در صورت PASS این مرحله، نسخه‌های زیر Freeze می‌شوند:
</p>

<ul>
<li>Extraction Prompt: <code>claim_extraction_prompt_v2</code></li>
<li>Semantic Validator: <code>claim_semantic_fidelity_v3</code></li>
<li>Exact Evidence Grounding: اجباری؛</li>
<li>Automatic Semantic Repair: ممنوع؛</li>
<li>Deterministic Critical-Risk Guard: اجباری؛</li>
<li>Runtime Model: <code>Qwen3-30B-A3B</code></li>
</ul>

<h3>Artifactهای این مرحله</h3>

<ul>
<li><code>claim_extraction_validation_contract_v1.json</code></li>
<li><code>claim_extraction_pilot_final_disposition_v1.jsonl</code></li>
<li><code>claim_extraction_pilot_validated_v1.jsonl</code></li>
<li><code>claim_extraction_pilot_final_report_v1.json</code></li>
</ul>

<h3>معیار PASS</h3>

<ul>
<li>تمام ۵۷ Raw Claim دقیقاً یک Final Disposition داشته باشند؛</li>
<li>هیچ REPAIR به‌صورت خودکار Validate نشود؛</li>
<li>هیچ Claim دارای Critical Risk Signal وارد Pilot Validated نشود؛</li>
<li>تمام Evidence Linkها مجدداً با Accepted Corpus تطبیق دقیق داشته باشند؛</li>
<li>Validated Pilot حداقل از هر سه Source Type Claim داشته باشد؛</li>
<li>Artifactها بدون Overwrite ناامن منتشر شوند؛</li>
<li>تمام Artifactها دوباره از Disk Validate شوند؛</li>
<li>هیچ API Call انجام نشود؛</li>
<li>در پایان Contract برای Full Registry به وضعیت Frozen برسد.</li>
</ul>

</div>

```python
# Code Cell 15

assert (
    pilot_semantic_v3_status
    == "PASS"
), (
    "Code Cell 14 must PASS before "
    "freezing the Claim Extraction Contract."
)


FINAL_VALIDATION_CONTRACT_VERSION = (
    "claim_extraction_validation_contract_v1"
)

FINAL_PILOT_DISPOSITION_VERSION = (
    "claim_extraction_pilot_final_disposition_v1"
)


CRITICAL_RISK_SIGNALS = {
    "FRAGMENT_OR_DISCOURSE_START",
    "SHORT_NUMERIC_CLAIM",
    "POSSIBLE_LIST_HEADER",
    "POSSIBLE_SOURCE_COMMENTARY",
    "SOURCE_FIRST_PERSON",
    "POSSIBLE_QUESTIONNAIRE_OR_SELF_REPORT",
    "POSSIBLE_FIRST_PERSON_GENERALIZATION",
}


validation_contract_path = (
    registry_artifact_root
    / "claim_extraction_validation_contract_v1.json"
)

pilot_final_disposition_path = (
    registry_artifact_root
    / "claim_extraction_pilot_final_disposition_v1.jsonl"
)

pilot_final_validated_path = (
    registry_artifact_root
    / "claim_extraction_pilot_validated_v1.jsonl"
)

pilot_final_report_path = (
    registry_artifact_root
    / "claim_extraction_pilot_final_report_v1.json"
)


# ---------------------------------------------------------
# Reload all authoritative Pilot-v2 / Semantic-v3
# artifacts independently from disk.
# ---------------------------------------------------------

final_raw_claims = load_jsonl(
    pilot_v2_raw_path
)

final_evidence_links = load_jsonl(
    pilot_v2_evidence_path
)

final_semantic_v3 = load_jsonl(
    semantic_v3_adjudication_path
)

final_semantic_v3_report = load_json(
    semantic_v3_report_path
)


assert (
    len(
        final_raw_claims
    )
    == 57
)


assert (
    len(
        final_semantic_v3
    )
    == 57
)


final_raw_claim_by_id = {
    item[
        "raw_claim_id"
    ]: item
    for item in (
        final_raw_claims
    )
}


final_semantic_by_id = {
    item[
        "raw_claim_id"
    ]: item
    for item in (
        final_semantic_v3
    )
}


assert (
    len(
        final_raw_claim_by_id
    )
    == 57
)


assert (
    len(
        final_semantic_by_id
    )
    == 57
)


assert (
    set(
        final_raw_claim_by_id
    )
    == set(
        final_semantic_by_id
    )
)


# ---------------------------------------------------------
# Revalidate all exact Evidence directly against
# the frozen Accepted Corpus.
# ---------------------------------------------------------

final_evidence_by_claim = defaultdict(
    list
)


for evidence in (
    final_evidence_links
):

    raw_claim_id = evidence[
        "raw_claim_id"
    ]


    assert (
        raw_claim_id
        in final_raw_claim_by_id
    )


    document_id = evidence[
        "document_id"
    ]


    assert (
        document_id
        in accepted_records_by_id
    )


    source_text = (
        accepted_records_by_id[
            document_id
        ][
            canonical_text_field
        ]
    )


    char_start = (
        evidence[
            "locator"
        ][
            "text_char_start"
        ]
    )

    char_end = (
        evidence[
            "locator"
        ][
            "text_char_end"
        ]
    )

    quote = evidence[
        "evidence_quote_exact"
    ]


    assert (
        source_text[
            char_start:
            char_end
        ]
        == quote
    ), (
        "Pilot Evidence no longer matches "
        f"Accepted Corpus: {evidence['evidence_id']}"
    )


    final_evidence_by_claim[
        raw_claim_id
    ].append(
        evidence
    )


for raw_claim_id in (
    final_raw_claim_by_id
):

    assert (
        final_evidence_by_claim[
            raw_claim_id
        ]
    ), (
        "Raw Claim has no Evidence: "
        f"{raw_claim_id}"
    )


# ---------------------------------------------------------
# Apply the frozen conservative disposition policy.
# No API and no semantic rewriting.
# ---------------------------------------------------------

final_dispositions = []

pilot_final_validated_claims = []


for raw_claim_id in sorted(
    final_raw_claim_by_id
):

    claim = (
        final_raw_claim_by_id[
            raw_claim_id
        ]
    )

    judgment = (
        final_semantic_by_id[
            raw_claim_id
        ]
    )


    semantic_decision = (
        judgment[
            "decision"
        ]
    )


    risk_signals = set(
        judgment.get(
            "fidelity_risk_signals",
            []
        )
    )


    triggered_critical_risks = sorted(
        risk_signals
        & CRITICAL_RISK_SIGNALS
    )


    if (
        semantic_decision
        == "REJECT"
    ):

        final_disposition = (
            "REJECT_SEMANTIC"
        )


    elif (
        semantic_decision
        == "REPAIR"
    ):

        final_disposition = (
            "HOLD_REPAIR_REQUIRED"
        )


    else:

        assert (
            semantic_decision
            == "ACCEPT"
        )


        if triggered_critical_risks:

            final_disposition = (
                "HOLD_CRITICAL_RISK"
            )

        else:

            final_disposition = (
                "VALIDATED"
            )


    evidence_items = (
        final_evidence_by_claim[
            raw_claim_id
        ]
    )


    source_types = sorted({
        item[
            "source_type"
        ]
        for item in (
            evidence_items
        )
    })


    assert (
        len(
            source_types
        )
        == 1
    )


    disposition_record = {
        "raw_claim_id": (
            raw_claim_id
        ),

        "final_disposition": (
            final_disposition
        ),

        "semantic_v3_decision": (
            semantic_decision
        ),

        "semantic_v3_source_statement_mode": (
            judgment[
                "source_statement_mode"
            ]
        ),

        "semantic_v3_issues": (
            judgment[
                "issues"
            ]
        ),

        "fidelity_risk_signals": sorted(
            risk_signals
        ),

        "triggered_critical_risk_signals": (
            triggered_critical_risks
        ),

        "claim_text_fa": (
            claim[
                "claim_text_fa"
            ]
        ),

        "claim_type": (
            claim[
                "claim_type"
            ]
        ),

        "source_type": (
            source_types[0]
        ),

        "authority_class": (
            claim[
                "authority_class"
            ]
        ),

        "claim_discovery_role": (
            claim[
                "claim_discovery_role"
            ]
        ),

        "evidence_ids": sorted(
            item[
                "evidence_id"
            ]
            for item in (
                evidence_items
            )
        ),

        "validation_contract_version": (
            FINAL_VALIDATION_CONTRACT_VERSION
        ),
    }


    final_dispositions.append(
        disposition_record
    )


    if (
        final_disposition
        == "VALIDATED"
    ):

        validated_claim = dict(
            claim
        )


        validated_claim[
            "registry_validation_status"
        ] = (
            "VALIDATED"
        )

        validated_claim[
            "validation_contract_version"
        ] = (
            FINAL_VALIDATION_CONTRACT_VERSION
        )

        validated_claim[
            "semantic_validation_version"
        ] = (
            SEMANTIC_V3_PROMPT_VERSION
        )

        validated_claim[
            "semantic_source_statement_mode"
        ] = (
            judgment[
                "source_statement_mode"
            ]
        )

        validated_claim[
            "semantic_issues"
        ] = []


        pilot_final_validated_claims.append(
            validated_claim
        )


# ---------------------------------------------------------
# Global disposition validation.
# ---------------------------------------------------------

final_disposition_ids = [
    item[
        "raw_claim_id"
    ]
    for item in (
        final_dispositions
    )
]


assert (
    len(
        final_disposition_ids
    )
    == 57
)


assert (
    len(
        final_disposition_ids
    )
    == len(
        set(
            final_disposition_ids
        )
    )
)


assert (
    set(
        final_disposition_ids
    )
    == set(
        final_raw_claim_by_id
    )
)


final_disposition_counts = Counter(
    item[
        "final_disposition"
    ]
    for item in (
        final_dispositions
    )
)


# No LLM REPAIR is automatically published.
assert not any(
    item[
        "final_disposition"
    ]
    == "VALIDATED"
    and item[
        "semantic_v3_decision"
    ]
    == "REPAIR"
    for item in (
        final_dispositions
    )
)


# No Critical Risk may enter validated output.
assert not any(
    item[
        "final_disposition"
    ]
    == "VALIDATED"
    and item[
        "triggered_critical_risk_signals"
    ]
    for item in (
        final_dispositions
    )
)


# Explicitly ensure every accepted critical-risk
# claim is held rather than auto-published.
critical_risk_accepts = [
    item
    for item in (
        final_dispositions
    )
    if (
        item[
            "semantic_v3_decision"
        ]
        == "ACCEPT"
        and item[
            "triggered_critical_risk_signals"
        ]
    )
]


assert all(
    item[
        "final_disposition"
    ]
    == "HOLD_CRITICAL_RISK"
    for item in (
        critical_risk_accepts
    )
)


# ---------------------------------------------------------
# Validate source coverage of the final Pilot set.
# ---------------------------------------------------------

validated_pilot_claim_ids = {
    item[
        "raw_claim_id"
    ]
    for item in (
        pilot_final_validated_claims
    )
}


validated_counts_by_source = Counter()

validated_counts_by_role = Counter()

validated_counts_by_type = Counter()


for claim in (
    pilot_final_validated_claims
):

    evidence_items = (
        final_evidence_by_claim[
            claim[
                "raw_claim_id"
            ]
        ]
    )


    source_type = (
        evidence_items[0][
            "source_type"
        ]
    )


    validated_counts_by_source[
        source_type
    ] += 1


    validated_counts_by_role[
        claim[
            "claim_discovery_role"
        ]
    ] += 1


    validated_counts_by_type[
        claim[
            "claim_type"
        ]
    ] += 1


for source_type in [
    "book",
    "transcript",
    "website",
]:

    assert (
        validated_counts_by_source[
            source_type
        ]
        > 0
    ), (
        "Final validated Pilot lost "
        f"all {source_type} Claims."
    )


# ---------------------------------------------------------
# Frozen Full Registry validation contract.
# ---------------------------------------------------------

validation_contract = {
    "contract_version": (
        FINAL_VALIDATION_CONTRACT_VERSION
    ),

    "status": (
        "FROZEN_FOR_FULL_REGISTRY"
    ),

    "runtime_model": (
        pilot_chat_model
    ),

    "source_population_contract": (
        "claim_extraction_prompt_window_contract_v1"
    ),

    "raw_extraction": {
        "prompt_version": (
            PILOT_V2_PROMPT_VERSION
        ),

        "temperature": (
            PILOT_V2_TEMPERATURE
        ),

        "max_tokens": (
            PILOT_V2_MAX_TOKENS
        ),

        "max_claims_per_window": (
            MAX_CLAIMS_PER_WINDOW
        ),

        "exact_evidence_required": True,

        "adaptive_retry": True,

        "automatic_claim_repair": False,
    },

    "semantic_validation": {
        "prompt_version": (
            SEMANTIC_V3_PROMPT_VERSION
        ),

        "temperature": (
            SEMANTIC_V3_TEMPERATURE
        ),

        "max_tokens": (
            SEMANTIC_V3_MAX_TOKENS
        ),

        "registered_evidence_only": True,

        "surrounding_context_is_evidence": False,

        "automatic_repair_publish_allowed": False,
    },

    "final_disposition_policy": {
        "semantic_reject": (
            "REJECT_SEMANTIC"
        ),

        "semantic_repair": (
            "HOLD_REPAIR_REQUIRED"
        ),

        "semantic_accept_with_critical_risk": (
            "HOLD_CRITICAL_RISK"
        ),

        "semantic_accept_without_critical_risk": (
            "VALIDATED"
        ),

        "critical_risk_signals": sorted(
            CRITICAL_RISK_SIGNALS
        ),
    },

    "registry_preservation_policy": {
        "raw_claims_preserved": True,
        "rejected_raw_claims_deleted": False,
        "held_raw_claims_deleted": False,
        "validated_claims_are_separate_artifact": True,
    },

    "full_registry_artifacts": {
        "raw_claims": (
            "claims_raw_v1.jsonl"
        ),

        "validated_claims": (
            "claims_validated_v1.jsonl"
        ),

        "evidence_links": (
            "claim_evidence_links_v1.jsonl"
        ),

        "raw_extraction_report": (
            "claim_extraction_full_report_v1.json"
        ),

        "validation_report": (
            "claim_validation_full_report_v1.json"
        ),
    },

    "execution_policy": {
        "batch_resumable": True,
        "validated_cache_required": True,
        "failed_response_cache_allowed": False,
        "overwrite_existing_artifacts": False,
        "disk_revalidation_required": True,
    },

    "source_artifact_hashes": {
        "prompt_window_manifest": (
            file_sha256(
                prompt_window_manifest_path
            )
        ),

        "pilot_v2_raw_claims": (
            file_sha256(
                pilot_v2_raw_path
            )
        ),

        "pilot_v2_evidence": (
            file_sha256(
                pilot_v2_evidence_path
            )
        ),

        "pilot_semantic_v3": (
            file_sha256(
                semantic_v3_adjudication_path
            )
        ),
    },

    "full_registry_extraction_started": (
        False
    ),
}


validation_contract_text = (
    json.dumps(
        validation_contract,
        ensure_ascii=False,
        indent=2,
        sort_keys=True,
    )
    + "\n"
)


validation_contract_sha256 = (
    hashlib.sha256(
        validation_contract_text.encode(
            "utf-8"
        )
    ).hexdigest()
)


# ---------------------------------------------------------
# Pilot final artifacts.
# ---------------------------------------------------------

final_dispositions = sorted(
    final_dispositions,
    key=lambda item: (
        item[
            "raw_claim_id"
        ]
    ),
)


pilot_final_validated_claims = sorted(
    pilot_final_validated_claims,
    key=lambda item: (
        item[
            "raw_claim_id"
        ]
    ),
)


pilot_final_disposition_text = "".join(
    json.dumps(
        item,
        ensure_ascii=False,
        sort_keys=True,
    )
    + "\n"
    for item in (
        final_dispositions
    )
)


pilot_final_validated_text = "".join(
    json.dumps(
        item,
        ensure_ascii=False,
        sort_keys=True,
    )
    + "\n"
    for item in (
        pilot_final_validated_claims
    )
)


pilot_final_disposition_sha256 = (
    hashlib.sha256(
        pilot_final_disposition_text.encode(
            "utf-8"
        )
    ).hexdigest()
)


pilot_final_validated_sha256 = (
    hashlib.sha256(
        pilot_final_validated_text.encode(
            "utf-8"
        )
    ).hexdigest()
)


pilot_final_report = {
    "report_version": (
        "claim_extraction_pilot_final_report_v1"
    ),

    "raw_claims": (
        len(
            final_raw_claims
        )
    ),

    "semantic_v3_decisions": (
        final_semantic_v3_report[
            "decision_counts"
        ]
    ),

    "final_dispositions": dict(
        sorted(
            final_disposition_counts.items()
        )
    ),

    "critical_risk_accepts_held": (
        len(
            critical_risk_accepts
        )
    ),

    "validated_claims": (
        len(
            pilot_final_validated_claims
        )
    ),

    "validated_by_source_type": dict(
        sorted(
            validated_counts_by_source.items()
        )
    ),

    "validated_by_claim_discovery_role": dict(
        sorted(
            validated_counts_by_role.items()
        )
    ),

    "validated_by_claim_type": dict(
        sorted(
            validated_counts_by_type.items()
        )
    ),

    "automatic_repairs_published": 0,

    "evidence_revalidation_failures": 0,

    "full_registry_contract": {
        "status": (
            "FROZEN_FOR_FULL_REGISTRY"
        ),

        "contract_version": (
            FINAL_VALIDATION_CONTRACT_VERSION
        ),

        "contract_sha256": (
            validation_contract_sha256
        ),
    },

    "full_registry_extraction_started": (
        False
    ),

    "artifacts": {
        "final_disposition": (
            pilot_final_disposition_path.name
        ),

        "final_disposition_sha256": (
            pilot_final_disposition_sha256
        ),

        "validated_pilot": (
            pilot_final_validated_path.name
        ),

        "validated_pilot_sha256": (
            pilot_final_validated_sha256
        ),
    },
}


pilot_final_report_text = (
    json.dumps(
        pilot_final_report,
        ensure_ascii=False,
        indent=2,
        sort_keys=True,
    )
    + "\n"
)


# ---------------------------------------------------------
# No-overwrite publication.
# ---------------------------------------------------------

def publish_frozen_artifact(
    path,
    content,
):
    if path.exists():

        existing = path.read_text(
            encoding="utf-8"
        )


        assert (
            existing
            == content
        ), (
            "Existing frozen artifact differs "
            f"from staged content: {path.name}. "
            "Refusing overwrite."
        )


        return (
            "REUSED_EXISTING_IDENTICAL"
        )


    with path.open(
        "x",
        encoding="utf-8",
    ) as file:
        file.write(
            content
        )


    return (
        "CREATED_NEW"
    )


validation_contract_write_status = (
    publish_frozen_artifact(
        validation_contract_path,
        validation_contract_text,
    )
)


pilot_final_disposition_write_status = (
    publish_frozen_artifact(
        pilot_final_disposition_path,
        pilot_final_disposition_text,
    )
)


pilot_final_validated_write_status = (
    publish_frozen_artifact(
        pilot_final_validated_path,
        pilot_final_validated_text,
    )
)


pilot_final_report_write_status = (
    publish_frozen_artifact(
        pilot_final_report_path,
        pilot_final_report_text,
    )
)


# ---------------------------------------------------------
# Independent disk revalidation.
# ---------------------------------------------------------

assert (
    load_json(
        validation_contract_path
    )
    == validation_contract
)


assert (
    load_jsonl(
        pilot_final_disposition_path
    )
    == final_dispositions
)


assert (
    load_jsonl(
        pilot_final_validated_path
    )
    == pilot_final_validated_claims
)


assert (
    load_json(
        pilot_final_report_path
    )
    == pilot_final_report
)


assert (
    file_sha256(
        validation_contract_path
    )
    == validation_contract_sha256
)


assert (
    file_sha256(
        pilot_final_disposition_path
    )
    == pilot_final_disposition_sha256
)


assert (
    file_sha256(
        pilot_final_validated_path
    )
    == pilot_final_validated_sha256
)


claim_extraction_validation_contract_status = (
    "PASS"
)


full_registry_readiness = (
    "READY_FOR_FULL_EXTRACTION"
)


# ---------------------------------------------------------
# Readable result.
# ---------------------------------------------------------

print(
    "=== NB12i FINAL PILOT CONTRACT FREEZE ==="
)

print()
print(
    "Raw Pilot Claims:",
    len(
        final_raw_claims
    ),
)

print(
    "Semantic v3 decisions:",
    final_semantic_v3_report[
        "decision_counts"
    ],
)

print()
print(
    "Final dispositions:",
    dict(
        sorted(
            final_disposition_counts.items()
        )
    ),
)

print(
    "Critical-risk ACCEPTs moved to HOLD:",
    len(
        critical_risk_accepts
    ),
)

print(
    "Automatic REPAIR publications:",
    0,
)

print()
print(
    "Validated Pilot Claims:",
    len(
        pilot_final_validated_claims
    ),
)

print(
    "Validated by source type:",
    dict(
        sorted(
            validated_counts_by_source.items()
        )
    ),
)

print(
    "Validated by role:",
    dict(
        sorted(
            validated_counts_by_role.items()
        )
    ),
)


print()
print(
    "=== CRITICAL-RISK ACCEPTS HELD ==="
)


if not critical_risk_accepts:

    print(
        "  None"
    )


for item in (
    critical_risk_accepts[:15]
):

    print()
    print(
        "Raw Claim ID:",
        item[
            "raw_claim_id"
        ],
    )

    print(
        "Claim:",
        item[
            "claim_text_fa"
        ],
    )

    print(
        "Critical signals:",
        item[
            "triggered_critical_risk_signals"
        ],
    )


print()
print(
    "Validation Contract:",
    validation_contract_path.relative_to(
        project_root
    ),
)

print(
    "Contract SHA-256:",
    validation_contract_sha256,
)

print(
    "Contract write status:",
    validation_contract_write_status,
)


print()
print(
    "Pilot Final Disposition:",
    pilot_final_disposition_path.relative_to(
        project_root
    ),
)

print(
    "Write status:",
    pilot_final_disposition_write_status,
)


print()
print(
    "Pilot Validated Claims:",
    pilot_final_validated_path.relative_to(
        project_root
    ),
)

print(
    "Write status:",
    pilot_final_validated_write_status,
)


print()
print(
    "Pilot Final Report:",
    pilot_final_report_path.relative_to(
        project_root
    ),
)

print(
    "Write status:",
    pilot_final_report_write_status,
)


print()
print(
    "API call performed:",
    False,
)

print(
    "Extraction Prompt frozen:",
    PILOT_V2_PROMPT_VERSION,
)

print(
    "Semantic Validator frozen:",
    SEMANTIC_V3_PROMPT_VERSION,
)

print(
    "Automatic semantic repair:",
    False,
)


print()
print(
    "Full Registry readiness:",
    full_registry_readiness,
)

print(
    "Full Registry Extraction:",
    "NOT_STARTED",
)


print()
print(
    "CODE CELL 15 GATE:",
    claim_extraction_validation_contract_status,
)
```

از این مرحله به بعد Prompt استخراج، Semantic Contract و قواعد Evidence دیگر بر اساس Pilot تغییر نمی‌کنند.

هدف این مرحله اجرای Raw Claim Extraction روی تمام ۲٬۲۸۷ Prompt Window رسمی Claim Registry است.

<h3>Pipeline این مرحله</h3>

<pre><code>2,287 Frozen Prompt Windows
↓
claim_extraction_prompt_v2
↓
Qwen3-30B-A3B
↓
Exact Evidence Validation
↓
Full Raw Claim Cache
↓
claims_raw_v1.jsonl
+
claim_evidence_links_v1.jsonl
</code></pre>

<h3>این مرحله هنوز Semantic Validation کامل را اجرا نمی‌کند</h3>

<p>
هر Claim تولیدشده در این مرحله یک Raw Claim است.
تمام Evidenceها باید Exact Source Span باشند، اما Claim هنوز تا زمانی که Semantic Fidelity Gate کامل روی Full Population اجرا نشده وارد <code>claims_validated_v1.jsonl</code> نخواهد شد.
</p>

<h3>Batch و Resume Policy</h3>

<p>
برای جلوگیری از ازبین‌رفتن کار در صورت قطع API یا Kernel، تمام Windowها مستقل Cache می‌شوند.
</p>

<ul>
<li>در هر اجرای این Cell حداکثر ۲۵۰ Window پردازش می‌شود؛</li>
<li>Windowهای Cache‌شده دوباره API مصرف نمی‌کنند؛</li>
<li>Cache فقط بعد از Parse و Exact Evidence Validation کامل نوشته می‌شود؛</li>
<li>پاسخ Invalid یا Failure Cache نمی‌شود؛</li>
<li>Pilot v2 Cache برای همان ۱۵ Window اولیه، پس از Revalidation، قابل استفادهٔ مجدد است؛</li>
<li>هر اجرای مجدد از اولین Window پردازش‌نشده ادامه می‌دهد.</li>
</ul>

<p>
عدد ۲۵۰ فقط اندازهٔ Batch عملیاتی برای اجرای Resume-safe است و Context Limit یا محدودیت معنایی مدل محسوب نمی‌شود.
</p>

<h3>Publication Policy</h3>

<p>
تا زمانی که تمام ۲٬۲۸۷ Window دارای Cache معتبر نباشند، Artifact نهایی Raw Registry منتشر نمی‌شود.
</p>

<p>
در اجرای میانی وضعیت Cell برابر <code>IN_PROGRESS</code> خواهد بود.
پس از تکمیل تمام Windowها، Cacheها دوباره از Disk خوانده و Validate می‌شوند و سپس Artifactهای نهایی ساخته خواهند شد.
</p>

<h3>قواعد Integrity</h3>

<ul>
<li>هر Prompt Window دقیقاً یک Cache Bundle دارد، حتی اگر صفر Claim تولید کند؛</li>
<li>هر Raw Claim حداقل یک Evidence Link دارد؛</li>
<li>تمام Evidenceها مستقیماً با Accepted Corpus تطبیق داده می‌شوند؛</li>
<li>هیچ Duplicate Raw Claim ID مجاز نیست؛</li>
<li>هیچ Duplicate Evidence ID مجاز نیست؛</li>
<li>هیچ Artifact نهایی قبل از کامل‌شدن تمام Windowها منتشر نمی‌شود؛</li>
<li>هیچ Artifact موجود Overwrite نمی‌شود مگر محتوای آن دقیقاً یکسان باشد؛</li>
<li>تمام Artifactهای نهایی پس از Publication دوباره از Disk Validate می‌شوند.</li>
</ul>

<h3>معیار PASS نهایی</h3>

<ul>
<li>تمام ۲٬۲۸۷ Prompt Window پردازش شده باشند؛</li>
<li>هیچ Window Failure حل‌نشده باقی نماند؛</li>
<li>تمام Raw Claimها Exact Evidence داشته باشند؛</li>
<li>تمام Evidenceها با Accepted Corpus تطبیق دقیق داشته باشند؛</li>
<li>تمام Source Typeها در Raw Registry حضور داشته باشند؛</li>
<li>Artifactهای نهایی از Disk مجدداً Validate شوند؛</li>
<li>Semantic Full Validation هنوز به مرحلهٔ بعد واگذار شود.</li>
</ul>

</div>

```python
# Code Cell 16

assert (
    claim_extraction_validation_contract_status
    == "PASS"
), (
    "Code Cell 15 must PASS before "
    "Full Raw Claim Extraction."
)

assert (
    full_registry_readiness
    == "READY_FOR_FULL_EXTRACTION"
), (
    "Frozen Pilot Contract is not ready "
    "for Full Extraction."
)


FULL_EXTRACTION_VERSION = (
    "claim_extraction_full_v1"
)

FULL_EXTRACTION_CACHE_VERSION = (
    "claim_extraction_full_cache_v1"
)

FULL_WINDOWS_PER_RUN = 250

# Operational retry policy only.
# The frozen semantic/extraction contract is unchanged.
FULL_MAX_ATTEMPTS = 6


full_raw_path = (
    registry_artifact_root
    / "claims_raw_v1.jsonl"
)

full_evidence_path = (
    registry_artifact_root
    / "claim_evidence_links_v1.jsonl"
)

full_extraction_report_path = (
    registry_artifact_root
    / "claim_extraction_full_report_v1.json"
)

full_extraction_cache_root = (
    registry_artifact_root
    / "claim_extraction_full_cache_v1"
)


full_extraction_cache_root.mkdir(
    parents=True,
    exist_ok=True,
)


# ---------------------------------------------------------
# Frozen Contract checks.
# ---------------------------------------------------------

disk_validation_contract = load_json(
    validation_contract_path
)


assert (
    disk_validation_contract[
        "status"
    ]
    == "FROZEN_FOR_FULL_REGISTRY"
)


assert (
    disk_validation_contract[
        "raw_extraction"
    ][
        "prompt_version"
    ]
    == PILOT_V2_PROMPT_VERSION
)


assert (
    disk_validation_contract[
        "runtime_model"
    ]
    == pilot_chat_model
)


assert (
    file_sha256(
        validation_contract_path
    )
    == validation_contract_sha256
)


# ---------------------------------------------------------
# Cache helpers.
# ---------------------------------------------------------

def full_cache_path_for_window(
    prompt_window_id,
):
    return (
        full_extraction_cache_root
        / (
            prompt_window_id.replace(
                ":",
                "_",
            )
            + ".json"
        )
    )


def relabel_bundle_for_full_registry(
    bundle,
):
    relabeled = json.loads(
        json.dumps(
            bundle,
            ensure_ascii=False,
        )
    )


    for claim in (
        relabeled[
            "raw_claims"
        ]
    ):

        claim[
            "extraction_version"
        ] = (
            FULL_EXTRACTION_VERSION
        )

        claim[
            "extraction_prompt_version"
        ] = (
            PILOT_V2_PROMPT_VERSION
        )

        claim[
            "validation_contract_version"
        ] = (
            FINAL_VALIDATION_CONTRACT_VERSION
        )


    for evidence in (
        relabeled[
            "evidence_links"
        ]
    ):

        evidence[
            "extraction_version"
        ] = (
            FULL_EXTRACTION_VERSION
        )

        evidence[
            "extraction_prompt_version"
        ] = (
            PILOT_V2_PROMPT_VERSION
        )

        evidence[
            "validation_contract_version"
        ] = (
            FINAL_VALIDATION_CONTRACT_VERSION
        )


    return relabeled


def revalidate_full_bundle(
    window,
    bundle,
):
    if not isinstance(
        bundle,
        dict,
    ):
        raise ValueError(
            "Full Extraction bundle "
            "must be an object."
        )


    if (
        bundle.get(
            "prompt_window_id"
        )
        != window[
            "prompt_window_id"
        ]
    ):
        raise ValueError(
            "Bundle Prompt Window mismatch."
        )


    raw_claims = bundle.get(
        "raw_claims"
    )

    evidence_links = bundle.get(
        "evidence_links"
    )


    if not isinstance(
        raw_claims,
        list,
    ):
        raise ValueError(
            "Bundle raw_claims "
            "must be a list."
        )


    if not isinstance(
        evidence_links,
        list,
    ):
        raise ValueError(
            "Bundle evidence_links "
            "must be a list."
        )


    raw_claim_ids = [
        item[
            "raw_claim_id"
        ]
        for item in (
            raw_claims
        )
    ]


    if (
        len(
            raw_claim_ids
        )
        != len(
            set(
                raw_claim_ids
            )
        )
    ):
        raise ValueError(
            "Duplicate Raw Claim ID "
            "inside Full bundle."
        )


    evidence_ids = [
        item[
            "evidence_id"
        ]
        for item in (
            evidence_links
        )
    ]


    if (
        len(
            evidence_ids
        )
        != len(
            set(
                evidence_ids
            )
        )
    ):
        raise ValueError(
            "Duplicate Evidence ID "
            "inside Full bundle."
        )


    raw_claim_id_set = set(
        raw_claim_ids
    )


    evidence_claim_ids = {
        item[
            "raw_claim_id"
        ]
        for item in (
            evidence_links
        )
    }


    if not evidence_claim_ids.issubset(
        raw_claim_id_set
    ):
        raise ValueError(
            "Evidence references unknown "
            "Raw Claim in Full bundle."
        )


    evidence_count_by_claim = Counter(
        item[
            "raw_claim_id"
        ]
        for item in (
            evidence_links
        )
    )


    for claim in (
        raw_claims
    ):

        if (
            claim[
                "prompt_window_id"
            ]
            != window[
                "prompt_window_id"
            ]
        ):
            raise ValueError(
                "Raw Claim Prompt Window mismatch."
            )


        if (
            claim[
                "extraction_version"
            ]
            != FULL_EXTRACTION_VERSION
        ):
            raise ValueError(
                "Raw Claim extraction version "
                "is not Full Registry version."
            )


        if (
            claim[
                "extraction_prompt_version"
            ]
            != PILOT_V2_PROMPT_VERSION
        ):
            raise ValueError(
                "Raw Claim Prompt version mismatch."
            )


        if (
            evidence_count_by_claim[
                claim[
                    "raw_claim_id"
                ]
            ]
            < 1
        ):
            raise ValueError(
                "Raw Claim has no Evidence."
            )


    for evidence in (
        evidence_links
    ):

        if (
            evidence[
                "prompt_window_id"
            ]
            != window[
                "prompt_window_id"
            ]
        ):
            raise ValueError(
                "Evidence Prompt Window mismatch."
            )


        if (
            evidence[
                "extraction_version"
            ]
            != FULL_EXTRACTION_VERSION
        ):
            raise ValueError(
                "Evidence extraction version "
                "is not Full Registry version."
            )


        document_id = evidence[
            "document_id"
        ]


        if (
            document_id
            not in accepted_records_by_id
        ):
            raise ValueError(
                "Evidence references unknown "
                f"document_id={document_id!r}"
            )


        source_text = (
            accepted_records_by_id[
                document_id
            ][
                canonical_text_field
            ]
        )


        char_start = (
            evidence[
                "locator"
            ][
                "text_char_start"
            ]
        )

        char_end = (
            evidence[
                "locator"
            ][
                "text_char_end"
            ]
        )

        quote = evidence[
            "evidence_quote_exact"
        ]


        if (
            source_text[
                char_start:
                char_end
            ]
            != quote
        ):
            raise ValueError(
                "Full Evidence does not match "
                "Accepted Corpus exactly."
            )


    return True


def build_full_cache_payload(
    window,
    bundle,
    cache_origin,
):
    return {
        "cache_version": (
            FULL_EXTRACTION_CACHE_VERSION
        ),

        "extraction_version": (
            FULL_EXTRACTION_VERSION
        ),

        "prompt_version": (
            PILOT_V2_PROMPT_VERSION
        ),

        "validation_contract_version": (
            FINAL_VALIDATION_CONTRACT_VERSION
        ),

        "model": (
            pilot_chat_model
        ),

        "prompt_window_id": (
            window[
                "prompt_window_id"
            ]
        ),

        "cache_origin": (
            cache_origin
        ),

        "bundle": (
            bundle
        ),
    }


def validate_full_cache_payload(
    window,
    payload,
):
    if not isinstance(
        payload,
        dict,
    ):
        raise ValueError(
            "Full cache payload "
            "must be an object."
        )


    expected_values = {
        "cache_version": (
            FULL_EXTRACTION_CACHE_VERSION
        ),

        "extraction_version": (
            FULL_EXTRACTION_VERSION
        ),

        "prompt_version": (
            PILOT_V2_PROMPT_VERSION
        ),

        "validation_contract_version": (
            FINAL_VALIDATION_CONTRACT_VERSION
        ),

        "model": (
            pilot_chat_model
        ),

        "prompt_window_id": (
            window[
                "prompt_window_id"
            ]
        ),
    }


    for (
        field,
        expected,
    ) in expected_values.items():

        if (
            payload.get(
                field
            )
            != expected
        ):
            raise ValueError(
                "Full cache field mismatch: "
                f"{field}"
            )


    revalidate_full_bundle(
        window,
        payload[
            "bundle"
        ],
    )


    return True


def write_full_cache(
    cache_path,
    payload,
):
    cache_text = (
        json.dumps(
            payload,
            ensure_ascii=False,
            indent=2,
            sort_keys=True,
        )
        + "\n"
    )


    if cache_path.exists():

        existing_text = (
            cache_path.read_text(
                encoding="utf-8"
            )
        )


        assert (
            existing_text
            == cache_text
        ), (
            "Existing Full cache differs "
            "from staged cache: "
            f"{cache_path.name}"
        )


        return (
            "REUSED_EXISTING_IDENTICAL"
        )


    with cache_path.open(
        "x",
        encoding="utf-8",
    ) as file:
        file.write(
            cache_text
        )


    return (
        "CREATED_NEW"
    )


# ---------------------------------------------------------
# Adaptive recovery feedback.
#
# Attempts 1-3 preserve normal Prompt-v2 behavior.
# Attempts 4-6 become increasingly conservative.
#
# This does NOT relax Evidence rules.
# It only allows fewer Claims when exact grounding
# proves difficult.
# ---------------------------------------------------------

def build_full_retry_feedback(
    last_error,
    attempt,
    slice_map,
):
    if not last_error:
        return None


    eligible_slice_ids = [
        slice_id
        for (
            slice_id,
            item,
        ) in (
            slice_map.items()
        )
        if item[
            "evidence_allowed"
        ]
    ]


    feedback_parts = [
        (
            "The previous response failed "
            "deterministic validation."
        ),
        (
            "Validator error:\n"
            f"{last_error}"
        ),
        (
            "Use ONLY these evidence-eligible "
            "slice_ids:\n"
            + json.dumps(
                eligible_slice_ids,
                ensure_ascii=False,
            )
        ),
        (
            "Every quote_exact must be copied "
            "character-for-character from the "
            "cited eligible source slice."
        ),
        (
            "Never cite an overlap-only slice."
        ),
        (
            "Never invent, paraphrase, normalize, "
            "rewrite, or clean Evidence text."
        ),
        (
            "Preserve the exact Unicode characters, "
            "Persian/Arabic letter forms, digits, "
            "punctuation, spaces, and half-spaces "
            "from the source."
        ),
    ]


    # -----------------------------------------------------
    # Malformed / missing JSON.
    # -----------------------------------------------------

    if (
        "JSONDecodeError"
        in last_error
        or "No JSON object found"
        in last_error
    ):

        feedback_parts.append(
            "Return exactly one compact, valid "
            "JSON object and no surrounding text."
        )


    if (
        "content is not a string"
        in last_error
    ):

        feedback_parts.append(
            "Return the required JSON object "
            "directly as normal text content."
        )


    # -----------------------------------------------------
    # Exact Evidence recovery.
    #
    # Found=0:
    # quote does not exist verbatim.
    #
    # Found>1:
    # quote exists but is not unique.
    #
    # These require opposite recovery strategies.
    # -----------------------------------------------------

    if (
        "Evidence quote must appear exactly once"
        in last_error
    ):

        if "Found=0" in last_error:

            feedback_parts.extend([
                (
                    "EVIDENCE RECOVERY — FOUND ZERO:"
                ),
                (
                    "The previous quote_exact was "
                    "NOT present verbatim in the "
                    "cited slice."
                ),
                (
                    "Copy a new quote_exact directly "
                    "from the visible source slice."
                ),
                (
                    "Use a contiguous exact substring. "
                    "Do not reconstruct the sentence "
                    "from memory."
                ),
                (
                    "Do not change ی/ي, ک/ك, digits, "
                    "punctuation, whitespace, or "
                    "half-spaces."
                ),
                (
                    "Prefer a short but meaningful "
                    "contiguous source span that can "
                    "be copied exactly."
                ),
                (
                    "If you cannot identify a truly "
                    "verbatim Evidence span for a "
                    "Claim, OMIT that Claim."
                ),
            ])

        else:

            feedback_parts.extend([
                (
                    "EVIDENCE RECOVERY — DUPLICATE QUOTE:"
                ),
                (
                    "The previous quote_exact occurs "
                    "more than once in the cited slice."
                ),
                (
                    "DO NOT shorten the quote."
                ),
                (
                    "Replace it with a LONGER exact "
                    "verbatim span by including "
                    "distinctive neighboring words "
                    "before and/or after it."
                ),
                (
                    "The new quote_exact must identify "
                    "only one occurrence in the cited "
                    "eligible slice."
                ),
                (
                    "Prefer the full sentence, or a "
                    "longer distinctive contiguous "
                    "phrase, when necessary."
                ),
                (
                    "Do not reuse the previous "
                    "ambiguous quote_exact."
                ),
            ])


    if (
        "overlap-only slice"
        in last_error
    ):

        feedback_parts.extend([
            (
                "The previous Evidence used an "
                "overlap-only slice."
            ),
            (
                "Choose Evidence only from one of "
                "the eligible slice_ids listed above."
            ),
        ])


    # -----------------------------------------------------
    # Conservative recovery.
    # -----------------------------------------------------

    if attempt >= 4:

        feedback_parts.extend([
            (
                "RECOVERY MODE: precision and exact "
                "grounding are more important than "
                "Claim count."
            ),
            (
                "Return at most 2 Claims."
            ),
            (
                "Omit every Claim whose Evidence "
                "cannot satisfy the exact-quote rule."
            ),
        ])


    if attempt >= 6:

        feedback_parts.extend([
            (
                "FINAL RECOVERY MODE:"
            ),
            (
                "Return at most 1 strongly grounded "
                "Claim."
            ),
            (
                "Before returning it, verify that "
                "quote_exact is copied verbatim from "
                "an evidence-eligible slice and is "
                "specific enough to occur only once."
            ),
            (
                "If no Claim can satisfy all Exact "
                "Evidence requirements with certainty, "
                'return exactly {"claims": []}.'
            ),
        ])


    return "\n\n".join(
        feedback_parts
    )


# ---------------------------------------------------------
# Validate all previously written Full caches.
# ---------------------------------------------------------

full_window_by_id = {
    window[
        "prompt_window_id"
    ]: window
    for window in (
        prompt_windows
    )
}


assert (
    len(
        full_window_by_id
    )
    == len(
        prompt_windows
    )
)


existing_full_cache_count = 0


for window in (
    prompt_windows
):

    cache_path = (
        full_cache_path_for_window(
            window[
                "prompt_window_id"
            ]
        )
    )


    if not cache_path.exists():
        continue


    payload = load_json(
        cache_path
    )


    validate_full_cache_payload(
        window,
        payload,
    )


    existing_full_cache_count += 1


pending_windows = [
    window
    for window in (
        prompt_windows
    )
    if not full_cache_path_for_window(
        window[
            "prompt_window_id"
        ]
    ).exists()
]


windows_this_run = (
    pending_windows[
        :FULL_WINDOWS_PER_RUN
    ]
)


# ---------------------------------------------------------
# Full Raw Extraction batch.
# ---------------------------------------------------------

full_api_calls_this_run = 0

full_pilot_cache_reuses_this_run = 0

full_new_cache_files_this_run = 0

full_batch_failures = []


print(
    "=== NB12i FULL RAW CLAIM EXTRACTION ==="
)

print()
print(
    "Frozen Prompt version:",
    PILOT_V2_PROMPT_VERSION,
)

print(
    "Frozen Validation Contract:",
    FINAL_VALIDATION_CONTRACT_VERSION,
)

print(
    "Model:",
    pilot_chat_model,
)

print()
print(
    "Total Prompt Windows:",
    len(
        prompt_windows
    ),
)

print(
    "Validated Full caches before run:",
    existing_full_cache_count,
)

print(
    "Pending before run:",
    len(
        pending_windows
    ),
)

print(
    "Windows scheduled this run:",
    len(
        windows_this_run
    ),
)

print()


for run_index, window in enumerate(
    windows_this_run,
    1,
):

    window_id = (
        window[
            "prompt_window_id"
        ]
    )


    cache_path = (
        full_cache_path_for_window(
            window_id
        )
    )


    print(
        f"[{run_index:03d}/{len(windows_this_run):03d}]",
        window[
            "source_type"
        ],
        "| role:",
        window[
            "claim_discovery_role"
        ],
        "| chars:",
        window[
            "raw_text_chars"
        ],
        "|",
        window_id,
    )


    # -----------------------------------------------------
    # Reuse validated Pilot-v2 cache if available.
    # -----------------------------------------------------

    pilot_cache_path = (
        pilot_v2_cache_root
        / (
            window_id.replace(
                ":",
                "_",
            )
            + ".json"
        )
    )


    if pilot_cache_path.exists():

        pilot_cached = load_json(
            pilot_cache_path
        )


        assert (
            pilot_cached[
                "prompt_version"
            ]
            == PILOT_V2_PROMPT_VERSION
        )


        assert (
            pilot_cached[
                "model"
            ]
            == pilot_chat_model
        )


        assert (
            pilot_cached[
                "prompt_window_id"
            ]
            == window_id
        )


        bundle = (
            relabel_bundle_for_full_registry(
                pilot_cached[
                    "bundle"
                ]
            )
        )


        revalidate_full_bundle(
            window,
            bundle,
        )


        payload = (
            build_full_cache_payload(
                window,
                bundle,
                cache_origin=(
                    "REUSED_VALIDATED_PILOT_V2"
                ),
            )
        )


        write_full_cache(
            cache_path,
            payload,
        )


        full_pilot_cache_reuses_this_run += 1

        full_new_cache_files_this_run += 1


        print(
            "   cache source: PILOT_V2",
            "| claims:",
            len(
                bundle[
                    "raw_claims"
                ]
            ),
        )


        continue


    # -----------------------------------------------------
    # New model extraction.
    # -----------------------------------------------------

    (
        context_text,
        slice_map,
    ) = materialize_pilot_window(
        window
    )


    validated_bundle = None

    last_error = None

    attempt_errors = []

    cache_origin = (
        "FULL_EXTRACTION_API"
    )


    for attempt in range(
        1,
        FULL_MAX_ATTEMPTS + 1,
    ):

        retry_feedback = (
            build_full_retry_feedback(
                last_error,
                attempt,
                slice_map,
            )
        )


        (
            system_message,
            user_message,
        ) = build_claim_prompt_v2(
            window,
            context_text,
            slice_map,
            retry_feedback=(
                retry_feedback
            ),
        )


        try:

            content = (
                call_pilot_model_v2(
                    system_message,
                    user_message,
                )
            )


            full_api_calls_this_run += 1


            model_payload = (
                parse_model_json(
                    content
                )
            )


            bundle = (
                validate_and_build_pilot_bundle_v2(
                    window,
                    model_payload,
                    slice_map,
                )
            )


            validated_bundle = (
                relabel_bundle_for_full_registry(
                    bundle
                )
            )


            revalidate_full_bundle(
                window,
                validated_bundle,
            )


            break


        except Exception as exc:

            last_error = (
                f"{type(exc).__name__}: "
                f"{exc}"
            )


            attempt_errors.append(
                last_error
            )


            print(
                f"   attempt {attempt}:",
                "INVALID",
                "|",
                last_error,
            )


            if (
                attempt
                < FULL_MAX_ATTEMPTS
            ):

                time.sleep(
                    min(
                        3.0,
                        float(
                            attempt
                        ),
                    )
                )


    # -----------------------------------------------------
    # Deterministic terminal recovery.
    #
    # If the model repeatedly produces Claims whose
    # Evidence cannot satisfy the frozen exact-quote rule,
    # publishing an unsupported Claim is forbidden.
    #
    # Therefore, after exhausting all retries, a Window
    # with persistent exact-Evidence failures is closed
    # conservatively as a valid zero-Claim Window.
    #
    # API/network/auth failures do NOT qualify for this
    # fallback and remain pending.
    # -----------------------------------------------------

    if validated_bundle is None:

        exact_evidence_error_count = sum(
            (
                "Evidence quote must appear "
                "exactly once in its slice."
            )
            in error
            for error in (
                attempt_errors
            )
        )


        allowed_terminal_error_tokens = (
            "Evidence quote must appear "
            "exactly once in its slice.",
            "JSONDecodeError:",
            "No JSON object found",
            "model response content is not a string",
        )


        terminal_errors_are_model_output_only = all(
            any(
                token in error
                for token in (
                    allowed_terminal_error_tokens
                )
            )
            for error in (
                attempt_errors
            )
        )


        persistent_exact_evidence_failure = (
            len(
                attempt_errors
            )
            == FULL_MAX_ATTEMPTS
            and exact_evidence_error_count
            >= 4
            and terminal_errors_are_model_output_only
        )


        if (
            persistent_exact_evidence_failure
        ):

            validated_bundle = {
                "prompt_window_id": (
                    window_id
                ),

                "raw_claims": [],

                "evidence_links": [],
            }


            revalidate_full_bundle(
                window,
                validated_bundle,
            )


            cache_origin = (
                "DETERMINISTIC_ZERO_CLAIM_"
                "AFTER_PERSISTENT_EXACT_"
                "EVIDENCE_FAILURE"
            )


            print(
                "   deterministic fallback:",
                "VALID",
                "| claims:",
                0,
            )


        else:

            full_batch_failures.append({
                "prompt_window_id": (
                    window_id
                ),

                "source_type": (
                    window[
                        "source_type"
                    ]
                ),

                "last_error": (
                    last_error
                ),

                "attempt_errors": (
                    attempt_errors
                ),
            })


            print(
                "   result: PENDING_RETRY"
            )


            continue


    payload = (
        build_full_cache_payload(
            window,
            validated_bundle,
            cache_origin=(
                cache_origin
            ),
        )
    )


    write_full_cache(
        cache_path,
        payload,
    )


    full_new_cache_files_this_run += 1


    print(
        "   result: VALID",
        "| claims:",
        len(
            validated_bundle[
                "raw_claims"
            ]
        ),
    )


# ---------------------------------------------------------
# Successful Window caches are intentionally preserved
# even when one or more Windows still need another run.
# ---------------------------------------------------------

print()


# ---------------------------------------------------------
# Recount and independently validate every Full cache.
# ---------------------------------------------------------

validated_full_cache_count = 0


for window in (
    prompt_windows
):

    cache_path = (
        full_cache_path_for_window(
            window[
                "prompt_window_id"
            ]
        )
    )


    if not cache_path.exists():
        continue


    payload = load_json(
        cache_path
    )


    validate_full_cache_payload(
        window,
        payload,
    )


    validated_full_cache_count += 1


remaining_full_windows = (
    len(
        prompt_windows
    )
    - validated_full_cache_count
)


print(
    "=== FULL EXTRACTION PROGRESS ==="
)

print()
print(
    "API calls this run:",
    full_api_calls_this_run,
)

print(
    "Pilot-v2 cache reuses this run:",
    full_pilot_cache_reuses_this_run,
)

print(
    "New validated Full caches this run:",
    full_new_cache_files_this_run,
)

print()
print(
    "Validated Full caches:",
    validated_full_cache_count,
    "/",
    len(
        prompt_windows
    ),
)

print(
    "Remaining Prompt Windows:",
    remaining_full_windows,
)

print(
    "Unresolved Windows this run:",
    len(
        full_batch_failures
    ),
)


if full_batch_failures:

    print()
    print(
        "=== WINDOWS PENDING RETRY ==="
    )


    for failure in (
        full_batch_failures
    ):

        print(
            "  -",
            failure[
                "prompt_window_id"
            ],
            "|",
            failure[
                "source_type"
            ],
            "|",
            failure[
                "last_error"
            ],
        )


# ---------------------------------------------------------
# Intermediate state.
#
# No final Registry publication until every Window
# has a validated cache.
# ---------------------------------------------------------

if (
    remaining_full_windows
    > 0
):

    assert not any(
        path.exists()
        for path in [
            full_raw_path,
            full_evidence_path,
            full_extraction_report_path,
        ]
    ), (
        "Final Full Extraction artifacts already "
        "exist while Full cache is incomplete."
    )


    full_raw_extraction_status = (
        "IN_PROGRESS"
    )


    print()
    print(
        "Successful caches preserved:",
        True,
    )

    print(
        "Failed Windows remain pending:",
        bool(
            full_batch_failures
        ),
    )

    print(
        "Final Registry artifacts published:",
        False,
    )

    print(
        "Semantic Full Validation:",
        "NOT_STARTED",
    )


    print()
    print(
        "CODE CELL 16 GATE:",
        full_raw_extraction_status,
    )


# ---------------------------------------------------------
# Complete state.
#
# Only when ALL 2,287 Windows are cached do we aggregate
# and publish immutable Full Raw Registry artifacts.
# ---------------------------------------------------------

else:

    assert not (
        full_batch_failures
    ), (
        "All Windows appear cached but this run "
        "still reports unresolved failures."
    )


    all_full_bundles = []


    for window in (
        prompt_windows
    ):

        cache_path = (
            full_cache_path_for_window(
                window[
                    "prompt_window_id"
                ]
            )
        )


        assert cache_path.exists()


        payload = load_json(
            cache_path
        )


        validate_full_cache_payload(
            window,
            payload,
        )


        all_full_bundles.append(
            payload[
                "bundle"
            ]
        )


    assert (
        len(
            all_full_bundles
        )
        == len(
            prompt_windows
        )
    )


    full_raw_claims = sorted(
        [
            claim
            for bundle in (
                all_full_bundles
            )
            for claim in (
                bundle[
                    "raw_claims"
                ]
            )
        ],
        key=lambda item: (
            item[
                "raw_claim_id"
            ]
        ),
    )


    full_evidence_links = sorted(
        [
            evidence
            for bundle in (
                all_full_bundles
            )
            for evidence in (
                bundle[
                    "evidence_links"
                ]
            )
        ],
        key=lambda item: (
            item[
                "evidence_id"
            ]
        ),
    )


    full_raw_claim_ids = [
        item[
            "raw_claim_id"
        ]
        for item in (
            full_raw_claims
        )
    ]


    full_evidence_ids = [
        item[
            "evidence_id"
        ]
        for item in (
            full_evidence_links
        )
    ]


    assert (
        len(
            full_raw_claim_ids
        )
        == len(
            set(
                full_raw_claim_ids
            )
        )
    ), (
        "Duplicate Raw Claim IDs detected "
        "in Full Registry aggregation."
    )


    assert (
        len(
            full_evidence_ids
        )
        == len(
            set(
                full_evidence_ids
            )
        )
    ), (
        "Duplicate Evidence IDs detected "
        "in Full Registry aggregation."
    )


    full_evidence_by_claim = Counter(
        item[
            "raw_claim_id"
        ]
        for item in (
            full_evidence_links
        )
    )


    assert all(
        full_evidence_by_claim[
            raw_claim_id
        ]
        >= 1
        for raw_claim_id in (
            full_raw_claim_ids
        )
    )


    full_claims_by_source = Counter()

    full_claims_by_role = Counter()

    full_claims_by_type = Counter()

    full_claims_by_window = Counter()


    for claim in (
        full_raw_claims
    ):

        window = (
            full_window_by_id[
                claim[
                    "prompt_window_id"
                ]
            ]
        )


        full_claims_by_source[
            window[
                "source_type"
            ]
        ] += 1


        full_claims_by_role[
            claim[
                "claim_discovery_role"
            ]
        ] += 1


        full_claims_by_type[
            claim[
                "claim_type"
            ]
        ] += 1


        full_claims_by_window[
            claim[
                "prompt_window_id"
            ]
        ] += 1


    zero_claim_windows = [
        window[
            "prompt_window_id"
        ]
        for window in (
            prompt_windows
        )
        if (
            full_claims_by_window[
                window[
                    "prompt_window_id"
                ]
            ]
            == 0
        )
    ]


    for source_type in [
        "book",
        "transcript",
        "website",
    ]:

        assert (
            full_claims_by_source[
                source_type
            ]
            > 0
        )


    # -----------------------------------------------------
    # Final Evidence validation directly against Corpus.
    # -----------------------------------------------------

    full_evidence_validation_failures = []


    for evidence in (
        full_evidence_links
    ):

        document_id = (
            evidence[
                "document_id"
            ]
        )

        source_text = (
            accepted_records_by_id[
                document_id
            ][
                canonical_text_field
            ]
        )

        char_start = (
            evidence[
                "locator"
            ][
                "text_char_start"
            ]
        )

        char_end = (
            evidence[
                "locator"
            ][
                "text_char_end"
            ]
        )

        quote = (
            evidence[
                "evidence_quote_exact"
            ]
        )


        if (
            source_text[
                char_start:
                char_end
            ]
            != quote
        ):

            full_evidence_validation_failures.append(
                evidence[
                    "evidence_id"
                ]
            )


    assert not (
        full_evidence_validation_failures
    ), (
        "Full Registry contains Evidence "
        "that no longer matches Accepted Corpus."
    )


    # -----------------------------------------------------
    # Serialize final artifacts.
    # -----------------------------------------------------

    full_raw_text = "".join(
        json.dumps(
            item,
            ensure_ascii=False,
            sort_keys=True,
        )
        + "\n"
        for item in (
            full_raw_claims
        )
    )


    full_evidence_text = "".join(
        json.dumps(
            item,
            ensure_ascii=False,
            sort_keys=True,
        )
        + "\n"
        for item in (
            full_evidence_links
        )
    )


    full_raw_sha256 = hashlib.sha256(
        full_raw_text.encode(
            "utf-8"
        )
    ).hexdigest()


    full_evidence_sha256 = hashlib.sha256(
        full_evidence_text.encode(
            "utf-8"
        )
    ).hexdigest()


    full_extraction_report = {
        "report_version": (
            "claim_extraction_full_report_v1"
        ),

        "extraction_version": (
            FULL_EXTRACTION_VERSION
        ),

        "validation_contract_version": (
            FINAL_VALIDATION_CONTRACT_VERSION
        ),

        "validation_contract_sha256": (
            validation_contract_sha256
        ),

        "prompt_version": (
            PILOT_V2_PROMPT_VERSION
        ),

        "model": (
            pilot_chat_model
        ),

        "prompt_windows_total": (
            len(
                prompt_windows
            )
        ),

        "validated_cache_bundles": (
            validated_full_cache_count
        ),

        "raw_claims": (
            len(
                full_raw_claims
            )
        ),

        "evidence_links": (
            len(
                full_evidence_links
            )
        ),

        "zero_claim_windows": (
            len(
                zero_claim_windows
            )
        ),

        "claims_by_source_type": dict(
            sorted(
                full_claims_by_source.items()
            )
        ),

        "claims_by_claim_discovery_role": dict(
            sorted(
                full_claims_by_role.items()
            )
        ),

        "claims_by_type": dict(
            sorted(
                full_claims_by_type.items()
            )
        ),

        "duplicate_raw_claim_ids": 0,

        "duplicate_evidence_ids": 0,

        "evidence_revalidation_failures": 0,

        "semantic_full_validation": (
            "NOT_STARTED"
        ),

        "validated_registry": (
            "NOT_YET_CREATED"
        ),

        "artifacts": {
            "raw_claims": (
                full_raw_path.name
            ),

            "raw_claims_sha256": (
                full_raw_sha256
            ),

            "evidence_links": (
                full_evidence_path.name
            ),

            "evidence_links_sha256": (
                full_evidence_sha256
            ),
        },
    }


    full_extraction_report_text = (
        json.dumps(
            full_extraction_report,
            ensure_ascii=False,
            indent=2,
            sort_keys=True,
        )
        + "\n"
    )


    # -----------------------------------------------------
    # Immutable publication.
    # -----------------------------------------------------

    full_raw_write_status = (
        publish_frozen_artifact(
            full_raw_path,
            full_raw_text,
        )
    )


    full_evidence_write_status = (
        publish_frozen_artifact(
            full_evidence_path,
            full_evidence_text,
        )
    )


    full_report_write_status = (
        publish_frozen_artifact(
            full_extraction_report_path,
            full_extraction_report_text,
        )
    )


    # -----------------------------------------------------
    # Independent disk validation.
    # -----------------------------------------------------

    assert (
        load_jsonl(
            full_raw_path
        )
        == full_raw_claims
    )


    assert (
        load_jsonl(
            full_evidence_path
        )
        == full_evidence_links
    )


    assert (
        load_json(
            full_extraction_report_path
        )
        == full_extraction_report
    )


    assert (
        file_sha256(
            full_raw_path
        )
        == full_raw_sha256
    )


    assert (
        file_sha256(
            full_evidence_path
        )
        == full_evidence_sha256
    )


    full_raw_extraction_status = (
        "PASS"
    )


    print()
    print(
        "=== NB12i FULL RAW CLAIM EXTRACTION RESULT ==="
    )

    print()
    print(
        "Prompt Windows processed:",
        len(
            prompt_windows
        ),
    )

    print(
        "Raw Claims:",
        len(
            full_raw_claims
        ),
    )

    print(
        "Evidence Links:",
        len(
            full_evidence_links
        ),
    )

    print(
        "Zero-Claim Windows:",
        len(
            zero_claim_windows
        ),
    )


    print()
    print(
        "Claims by source type:",
        dict(
            sorted(
                full_claims_by_source.items()
            )
        ),
    )

    print(
        "Claims by authority role:",
        dict(
            sorted(
                full_claims_by_role.items()
            )
        ),
    )

    print(
        "Claims by type:",
        dict(
            sorted(
                full_claims_by_type.items()
            )
        ),
    )


    print()
    print(
        "Duplicate Raw Claim IDs:",
        0,
    )

    print(
        "Duplicate Evidence IDs:",
        0,
    )

    print(
        "Evidence revalidation failures:",
        0,
    )


    print()
    print(
        "Raw Claims Artifact:",
        full_raw_path.relative_to(
            project_root
        ),
    )

    print(
        "Raw Claims SHA-256:",
        full_raw_sha256,
    )

    print(
        "Write status:",
        full_raw_write_status,
    )


    print()
    print(
        "Evidence Artifact:",
        full_evidence_path.relative_to(
            project_root
        ),
    )

    print(
        "Evidence SHA-256:",
        full_evidence_sha256,
    )

    print(
        "Write status:",
        full_evidence_write_status,
    )


    print()
    print(
        "Full Extraction Report:",
        full_extraction_report_path.relative_to(
            project_root
        ),
    )

    print(
        "Write status:",
        full_report_write_status,
    )


    print()
    print(
        "Semantic Full Validation:",
        "NOT_STARTED",
    )

    print(
        "Validated Registry:",
        "NOT_YET_CREATED",
    )


    print()
    print(
        "CODE CELL 16 GATE:",
        full_raw_extraction_status,
    )
```

<h2>آماده‌سازی Full Semantic Validation با Validator فریز‌شده</h2>

<p>
استخراج کامل Raw Claimها با موفقیت پایان یافته است. در این مرحله باید Validator معنایی فریز‌شده
<code>claim_semantic_fidelity_v3</code>
بدون تغییر سیاست یا Prompt روی کل Registry اجرا شود.
</p>

<p>
پیش از شروع فراخوانی‌های انبوه API، این مرحله فقط اتصال دقیق بین خروجی کامل استخراج و Validator فریز‌شده قبلی را بررسی می‌کند تا نام تابع‌ها، Schema خروجی و Artifactهای مرجع از وضعیت واقعی Notebook و Disk خوانده شوند و هیچ جزئیاتی حدس زده نشود.
</p>

<ul>
<li>ورودی اصلی: <code>claims_raw_v1.jsonl</code> و <code>claim_evidence_links_v1.jsonl</code></li>
<li>تعداد مورد انتظار Raw Claim بر اساس خروجی پذیرفته‌شده مرحله قبل: <code>6237</code></li>
<li>Semantic policy باید همان نسخه فریز‌شده v3 باقی بماند.</li>
<li>در این مرحله هیچ API Call انجام نمی‌شود.</li>
<li>هیچ فایل یا Artifact جدیدی ساخته یا بازنویسی نمی‌شود.</li>
</ul>

<p>
Gate زمانی PASS است که Artifactهای Full Extraction با Hash ثبت‌شده دوباره از Disk تأیید شوند، Artifactهای Pilot Semantic v3 قابل خواندن باشند، Schema واقعی آن‌ها گزارش شود و Symbolهای Semantic/Fidelity موجود در Runtime بدون تغییر نمایش داده شوند.
</p>

</div>

```python
# Code Cell 17

assert (
    full_raw_extraction_status
    == "PASS"
), (
    "Code Cell 16 must PASS before "
    "Full Semantic Validation handoff."
)


EXPECTED_FULL_RAW_SHA256 = (
    "fff018199d96fab3a59729527c56a0c4"
    "ba0aa61d9d3a133b3d48645b62362d02"
)

EXPECTED_FULL_EVIDENCE_SHA256 = (
    "1c47a493b2128b1fd489752378a3e5d4"
    "dc7e53b7379b03674abe37d9a77cc9ba"
)

EXPECTED_FULL_RAW_CLAIMS = 6237
EXPECTED_FULL_EVIDENCE_LINKS = 6396


pilot_semantic_v3_path = (
    registry_artifact_root
    / (
        "claim_extraction_pilot_"
        "semantic_adjudication_v3.jsonl"
    )
)

pilot_semantic_v3_report_path = (
    registry_artifact_root
    / (
        "claim_extraction_pilot_"
        "semantic_report_v3.json"
    )
)


required_semantic_handoff_paths = [
    full_raw_path,
    full_evidence_path,
    full_extraction_report_path,
    validation_contract_path,
    pilot_semantic_v3_path,
    pilot_semantic_v3_report_path,
]


missing_semantic_handoff_paths = [
    path
    for path in required_semantic_handoff_paths
    if not path.exists()
]


assert not missing_semantic_handoff_paths, (
    "Missing Semantic handoff artifacts:\n"
    + "\n".join(
        str(path)
        for path in missing_semantic_handoff_paths
    )
)


disk_full_raw_claims = load_jsonl(
    full_raw_path
)

disk_full_evidence_links = load_jsonl(
    full_evidence_path
)

disk_full_extraction_report = load_json(
    full_extraction_report_path
)

disk_validation_contract = load_json(
    validation_contract_path
)

disk_pilot_semantic_v3 = load_jsonl(
    pilot_semantic_v3_path
)

disk_pilot_semantic_v3_report = load_json(
    pilot_semantic_v3_report_path
)


assert (
    len(
        disk_full_raw_claims
    )
    == EXPECTED_FULL_RAW_CLAIMS
), (
    "Full Raw Claim count changed."
)


assert (
    len(
        disk_full_evidence_links
    )
    == EXPECTED_FULL_EVIDENCE_LINKS
), (
    "Full Evidence Link count changed."
)


assert (
    file_sha256(
        full_raw_path
    )
    == EXPECTED_FULL_RAW_SHA256
), (
    "Full Raw Claims SHA-256 changed."
)


assert (
    file_sha256(
        full_evidence_path
    )
    == EXPECTED_FULL_EVIDENCE_SHA256
), (
    "Full Evidence SHA-256 changed."
)


assert (
    disk_full_extraction_report[
        "raw_claims"
    ]
    == EXPECTED_FULL_RAW_CLAIMS
)


assert (
    disk_full_extraction_report[
        "evidence_links"
    ]
    == EXPECTED_FULL_EVIDENCE_LINKS
)


assert (
    disk_full_extraction_report[
        "semantic_full_validation"
    ]
    == "NOT_STARTED"
)


assert (
    disk_full_extraction_report[
        "validated_registry"
    ]
    == "NOT_YET_CREATED"
)


assert (
    disk_pilot_semantic_v3
), (
    "Pilot Semantic v3 artifact is empty."
)


pilot_semantic_v3_keys = sorted(
    {
        key
        for row in disk_pilot_semantic_v3
        if isinstance(
            row,
            dict,
        )
        for key in row.keys()
    }
)


pilot_semantic_v3_first_record = (
    disk_pilot_semantic_v3[0]
)


assert isinstance(
    pilot_semantic_v3_first_record,
    dict,
), (
    "Pilot Semantic v3 records "
    "must be JSON objects."
)


semantic_runtime_symbols = []


for symbol_name, symbol_value in sorted(
    globals().items()
):

    normalized_name = (
        symbol_name.casefold()
    )

    if (
        "semantic"
        not in normalized_name
        and "fidelity"
        not in normalized_name
    ):
        continue


    semantic_runtime_symbols.append({
        "name": symbol_name,
        "type": type(
            symbol_value
        ).__name__,
        "callable": callable(
            symbol_value
        ),
    })


semantic_callable_symbols = [
    item
    for item in semantic_runtime_symbols
    if item[
        "callable"
    ]
]


semantic_v3_named_symbols = [
    item
    for item in semantic_runtime_symbols
    if (
        "v3"
        in item[
            "name"
        ].casefold()
        or "fidelity"
        in item[
            "name"
        ].casefold()
    )
]


assert semantic_runtime_symbols, (
    "No Semantic/Fidelity runtime symbols "
    "were found. Code Cells 14-15 may not "
    "have been executed in this kernel."
)


semantic_handoff_status = "PASS"


print(
    "=== NB12i FULL SEMANTIC VALIDATION HANDOFF ==="
)

print()
print(
    "Full Raw Claims:",
    len(
        disk_full_raw_claims
    ),
)

print(
    "Full Evidence Links:",
    len(
        disk_full_evidence_links
    ),
)

print(
    "Full Raw SHA-256:",
    file_sha256(
        full_raw_path
    ),
)

print(
    "Full Evidence SHA-256:",
    file_sha256(
        full_evidence_path
    ),
)


print()
print(
    "Pilot Semantic v3 records:",
    len(
        disk_pilot_semantic_v3
    ),
)

print(
    "Pilot Semantic v3 SHA-256:",
    file_sha256(
        pilot_semantic_v3_path
    ),
)

print(
    "Pilot Semantic v3 report SHA-256:",
    file_sha256(
        pilot_semantic_v3_report_path
    ),
)


print()
print(
    "Pilot Semantic v3 record keys:",
    pilot_semantic_v3_keys,
)


print()
print(
    "First Semantic v3 record:"
)

print(
    json.dumps(
        pilot_semantic_v3_first_record,
        ensure_ascii=False,
        indent=2,
        sort_keys=True,
    )
)


print()
print(
    "Semantic/Fidelity runtime symbols:",
    len(
        semantic_runtime_symbols
    ),
)


for item in semantic_runtime_symbols:

    print(
        "  -",
        item[
            "name"
        ],
        "| type:",
        item[
            "type"
        ],
        "| callable:",
        item[
            "callable"
        ],
    )


print()
print(
    "Callable Semantic/Fidelity symbols:",
    len(
        semantic_callable_symbols
    ),
)

print(
    "V3/Fidelity-named symbols:",
    len(
        semantic_v3_named_symbols
    ),
)


print()
print(
    "API calls performed:",
    0,
)

print(
    "Files modified:",
    False,
)


print()
print(
    "CODE CELL 17 GATE:",
    semantic_handoff_status,
)
```

<h2>اعتبارسنجی معنایی کامل Claim Registry</h2>

<p>
استخراج کامل Raw Claimها پایان یافته و اکنون تمام Claimهای خام باید با همان
<code>claim_semantic_fidelity_v3</code>
که در Pilot فریز و تأیید شده بود، از نظر وفاداری معنایی به Evidence بررسی شوند.
</p>

<p>
این مرحله هیچ Claim جدیدی استخراج نمی‌کند و Evidence را تغییر نمی‌دهد. برای هر Claim فقط تصمیم
<code>ACCEPT</code>،
<code>REPAIR</code>
یا
<code>REJECT</code>
تولید می‌شود. تصمیم‌های <code>REPAIR</code> در این مرحله به‌صورت خودکار وارد Registry نهایی نمی‌شوند.
</p>

<ul>
<li>ورودی: ۶۲۳۷ Raw Claim و ۶۳۹۶ Evidence Link فریز‌شده.</li>
<li>Validator: همان Semantic Fidelity v3 فریز‌شده.</li>
<li>۵۷ داوری Pilot فقط بعد از Re-validation مستقل reuse می‌شوند.</li>
<li>بقیه Claimها در Batchهای چهار‌تایی پردازش می‌شوند.</li>
<li>Cache هر Batch بلافاصله روی Disk ذخیره می‌شود و اجرای بعدی از همان نقطه ادامه می‌دهد.</li>
<li>وجود خطای موقت در چند Batch باعث از بین رفتن Batchهای موفق نمی‌شود.</li>
<li>Artifact نهایی Semantic فقط بعد از تکمیل تمام Batchها منتشر می‌شود.</li>
<li>اعمال Critical-Risk Guard و ساخت <code>claims_validated_v1.jsonl</code> در مرحله بعد انجام خواهد شد.</li>
</ul>

<p>
Gate این مرحله فقط زمانی PASS می‌شود که هر ۶۲۳۷ Claim یک Semantic Judgment معتبر داشته باشد، هیچ ID مفقود یا تکراری نباشد و Artifact نهایی دوباره مستقل از Disk تأیید شود.
</p>

```python
# Code Cell 18

from concurrent.futures import (
    ThreadPoolExecutor,
    as_completed,
)
import threading


assert (
    semantic_handoff_status
    == "PASS"
), (
    "Code Cell 17 must PASS before "
    "Full Semantic Validation."
)

assert (
    full_raw_extraction_status
    == "PASS"
), (
    "Full Raw Claim Extraction must PASS."
)


FULL_SEMANTIC_VERSION = (
    "claim_validation_full_semantic_v1"
)

FULL_SEMANTIC_CACHE_VERSION = (
    "claim_validation_full_semantic_cache_v1"
)


# ---------------------------------------------------------
# Parallel execution policy.
#
# Provider hard limit:
#   30 requests / minute
#
# Internal limit:
#   24 requests / minute
#
# This preserves ~20% operational headroom.
#
# Semantic Prompt, model, Batch size and validation
# contract are NOT changed.
# ---------------------------------------------------------

FULL_SEMANTIC_MAX_WORKERS = 6

FULL_SEMANTIC_REQUESTS_PER_MINUTE = 24

FULL_SEMANTIC_PROGRESS_EVERY = 10


assert (
    FULL_SEMANTIC_REQUESTS_PER_MINUTE
    < 30
), (
    "Internal request rate must remain "
    "below the provider limit."
)


full_semantic_cache_root = (
    registry_artifact_root
    / "claim_validation_full_semantic_cache_v1"
)

full_semantic_adjudication_path = (
    registry_artifact_root
    / "claim_validation_full_semantic_adjudication_v1.jsonl"
)

full_semantic_report_path = (
    registry_artifact_root
    / "claim_validation_full_semantic_report_v1.json"
)


full_semantic_cache_root.mkdir(
    parents=True,
    exist_ok=True,
)


# ---------------------------------------------------------
# Independently reload frozen Full Extraction artifacts.
# ---------------------------------------------------------

full_semantic_claims = load_jsonl(
    full_raw_path
)

full_semantic_evidence = load_jsonl(
    full_evidence_path
)


assert (
    len(
        full_semantic_claims
    )
    == EXPECTED_FULL_RAW_CLAIMS
)


assert (
    len(
        full_semantic_evidence
    )
    == EXPECTED_FULL_EVIDENCE_LINKS
)


assert (
    file_sha256(
        full_raw_path
    )
    == EXPECTED_FULL_RAW_SHA256
)


assert (
    file_sha256(
        full_evidence_path
    )
    == EXPECTED_FULL_EVIDENCE_SHA256
)


assert (
    SEMANTIC_V3_PROMPT_VERSION
    == "claim_semantic_fidelity_v3"
)


assert (
    disk_validation_contract[
        "semantic_validation"
    ][
        "prompt_version"
    ]
    == SEMANTIC_V3_PROMPT_VERSION
)


assert (
    disk_validation_contract[
        "runtime_model"
    ]
    == pilot_chat_model
)


# ---------------------------------------------------------
# Full Claim / Evidence indexes.
# ---------------------------------------------------------

full_semantic_claim_by_id = {
    item[
        "raw_claim_id"
    ]: item
    for item in (
        full_semantic_claims
    )
}


assert (
    len(
        full_semantic_claim_by_id
    )
    == len(
        full_semantic_claims
    )
), (
    "Duplicate raw_claim_id detected "
    "before Full Semantic Validation."
)


full_semantic_evidence_by_claim = defaultdict(
    list
)


for evidence in (
    full_semantic_evidence
):

    raw_claim_id = evidence[
        "raw_claim_id"
    ]


    assert (
        raw_claim_id
        in full_semantic_claim_by_id
    ), (
        "Evidence references unknown "
        f"Raw Claim: {raw_claim_id}"
    )


    full_semantic_evidence_by_claim[
        raw_claim_id
    ].append(
        evidence
    )


for raw_claim_id in (
    full_semantic_claim_by_id
):

    assert (
        full_semantic_evidence_by_claim[
            raw_claim_id
        ]
    ), (
        "Raw Claim has no registered Evidence: "
        f"{raw_claim_id}"
    )


# ---------------------------------------------------------
# Materialize Full Semantic v3 items using the frozen
# Pilot-v3 evidence/context/risk logic.
# ---------------------------------------------------------

full_semantic_items = []


for claim in sorted(
    full_semantic_claims,
    key=lambda item: (
        item[
            "raw_claim_id"
        ]
    ),
):

    raw_claim_id = claim[
        "raw_claim_id"
    ]


    evidence_items = [
        build_semantic_v3_evidence_item(
            evidence
        )
        for evidence in sorted(
            full_semantic_evidence_by_claim[
                raw_claim_id
            ],
            key=lambda item: (
                item[
                    "evidence_id"
                ]
            ),
        )
    ]


    evidence_quotes = [
        item[
            "quote_exact"
        ]
        for item in (
            evidence_items
        )
    ]


    surrounding_contexts = [
        item[
            "judge_only_surrounding_context"
        ]
        for item in (
            evidence_items
        )
    ]


    full_semantic_items.append({
        "raw_claim_id": (
            raw_claim_id
        ),

        "claim_text_fa": (
            claim[
                "claim_text_fa"
            ]
        ),

        "claim_type": (
            claim[
                "claim_type"
            ]
        ),

        "authority_class": (
            claim[
                "authority_class"
            ]
        ),

        "claim_discovery_role": (
            claim[
                "claim_discovery_role"
            ]
        ),

        "fidelity_risk_signals": (
            detect_fidelity_risk_signals(
                claim[
                    "claim_text_fa"
                ],
                evidence_quotes,
                surrounding_contexts,
            )
        ),

        "evidence": (
            evidence_items
        ),
    })


assert (
    len(
        full_semantic_items
    )
    == EXPECTED_FULL_RAW_CLAIMS
)


full_semantic_item_by_id = {
    item[
        "raw_claim_id"
    ]: item
    for item in (
        full_semantic_items
    )
}


assert (
    len(
        full_semantic_item_by_id
    )
    == EXPECTED_FULL_RAW_CLAIMS
)


# ---------------------------------------------------------
# Reuse Pilot Semantic v3 judgments only after independent
# validation against the current Full Registry.
# ---------------------------------------------------------

pilot_semantic_reused_by_id = {}


for pilot_judgment in (
    disk_pilot_semantic_v3
):

    raw_claim_id = (
        pilot_judgment[
            "raw_claim_id"
        ]
    )


    assert (
        raw_claim_id
        in full_semantic_item_by_id
    ), (
        "Pilot Semantic Claim is missing "
        "from Full Raw Registry: "
        f"{raw_claim_id}"
    )


    full_item = (
        full_semantic_item_by_id[
            raw_claim_id
        ]
    )


    revalidated = (
        validate_semantic_v3_payload(
            [full_item],
            {
                "judgments": [
                    pilot_judgment
                ]
            },
        )
    )


    assert (
        len(
            revalidated
        )
        == 1
    )


    assert (
        revalidated[0]
        == pilot_judgment
    ), (
        "Pilot Semantic judgment differs "
        "when independently revalidated "
        "against Full Registry evidence: "
        f"{raw_claim_id}"
    )


    pilot_semantic_reused_by_id[
        raw_claim_id
    ] = (
        pilot_judgment
    )


assert (
    len(
        pilot_semantic_reused_by_id
    )
    == len(
        disk_pilot_semantic_v3
    )
)


# ---------------------------------------------------------
# Claims still requiring Full Semantic API adjudication.
# ---------------------------------------------------------

full_semantic_api_items = [
    item
    for item in (
        full_semantic_items
    )
    if (
        item[
            "raw_claim_id"
        ]
        not in pilot_semantic_reused_by_id
    )
]


full_semantic_batches = [
    full_semantic_api_items[
        start:
        start
        + SEMANTIC_V3_BATCH_SIZE
    ]
    for start in range(
        0,
        len(
            full_semantic_api_items
        ),
        SEMANTIC_V3_BATCH_SIZE,
    )
]


assert all(
    1
    <= len(
        batch_items
    )
    <= SEMANTIC_V3_BATCH_SIZE
    for batch_items in (
        full_semantic_batches
    )
)


# ---------------------------------------------------------
# Full Semantic cache helpers.
# ---------------------------------------------------------

def full_semantic_cache_path_for_batch(
    batch_id,
):
    return (
        full_semantic_cache_root
        / (
            batch_id.replace(
                ":",
                "_",
            )
            + ".json"
        )
    )


def validate_full_semantic_cache(
    batch_items,
    payload,
):
    if not isinstance(
        payload,
        dict,
    ):
        raise ValueError(
            "Full Semantic cache payload "
            "must be an object."
        )


    batch_id = (
        semantic_v3_batch_id(
            batch_items
        )
    )


    expected_metadata = {
        "cache_version": (
            FULL_SEMANTIC_CACHE_VERSION
        ),

        "validation_version": (
            FULL_SEMANTIC_VERSION
        ),

        "batch_id": (
            batch_id
        ),

        "semantic_prompt_version": (
            SEMANTIC_V3_PROMPT_VERSION
        ),

        "model": (
            pilot_chat_model
        ),

        "source_raw_claims_sha256": (
            EXPECTED_FULL_RAW_SHA256
        ),

        "source_evidence_sha256": (
            EXPECTED_FULL_EVIDENCE_SHA256
        ),

        "validation_contract_sha256": (
            validation_contract_sha256
        ),
    }


    for (
        field,
        expected_value,
    ) in (
        expected_metadata.items()
    ):

        if (
            payload.get(
                field
            )
            != expected_value
        ):
            raise ValueError(
                "Full Semantic cache "
                f"metadata mismatch: {field}"
            )


    expected_claim_ids = [
        item[
            "raw_claim_id"
        ]
        for item in (
            batch_items
        )
    ]


    if (
        payload.get(
            "raw_claim_ids"
        )
        != expected_claim_ids
    ):
        raise ValueError(
            "Full Semantic cache Claim IDs "
            "do not match deterministic batch."
        )


    cached_judgments = (
        payload.get(
            "judgments"
        )
    )


    canonical_judgments = (
        validate_semantic_v3_payload(
            batch_items,
            {
                "judgments": (
                    cached_judgments
                )
            },
        )
    )


    if (
        canonical_judgments
        != cached_judgments
    ):
        raise ValueError(
            "Full Semantic cache judgments "
            "are not canonical."
        )


    return (
        canonical_judgments
    )


def write_full_semantic_cache(
    cache_path,
    payload,
):
    cache_text = (
        json.dumps(
            payload,
            ensure_ascii=False,
            indent=2,
            sort_keys=True,
        )
        + "\n"
    )


    if cache_path.exists():

        existing_text = (
            cache_path.read_text(
                encoding="utf-8"
            )
        )


        assert (
            existing_text
            == cache_text
        ), (
            "Existing Full Semantic cache "
            "differs from staged cache: "
            f"{cache_path.name}"
        )


        return (
            "REUSED_EXISTING_IDENTICAL"
        )


    with cache_path.open(
        "x",
        encoding="utf-8",
    ) as file:

        file.write(
            cache_text
        )


    return (
        "CREATED_NEW"
    )


# ---------------------------------------------------------
# Global thread-safe request-rate limiter.
#
# Request starts are intentionally spaced rather than
# burst in groups. This keeps the endpoint safely below
# the provider's fixed 30 request/minute limit.
# ---------------------------------------------------------

class FullSemanticRateLimiter:

    def __init__(
        self,
        requests_per_minute,
    ):
        assert (
            requests_per_minute
            > 0
        )

        self.interval_seconds = (
            60.0
            / float(
                requests_per_minute
            )
        )

        self.lock = threading.Lock()

        self.next_request_time = (
            time.monotonic()
        )


    def wait_for_slot(
        self,
    ):
        with self.lock:

            now = time.monotonic()

            wait_seconds = max(
                0.0,
                (
                    self.next_request_time
                    - now
                ),
            )


            if (
                wait_seconds
                > 0
            ):

                time.sleep(
                    wait_seconds
                )


            actual_start = (
                time.monotonic()
            )


            self.next_request_time = (
                max(
                    actual_start,
                    self.next_request_time,
                )
                + self.interval_seconds
            )


full_semantic_rate_limiter = (
    FullSemanticRateLimiter(
        FULL_SEMANTIC_REQUESTS_PER_MINUTE
    )
)


# ---------------------------------------------------------
# One independent Semantic Batch worker.
# ---------------------------------------------------------

def process_full_semantic_batch(
    batch_items,
):
    batch_id = (
        semantic_v3_batch_id(
            batch_items
        )
    )


    cache_path = (
        full_semantic_cache_path_for_batch(
            batch_id
        )
    )


    validated_judgments = None

    last_error = None

    attempt_errors = []

    api_calls = 0


    for attempt in range(
        1,
        SEMANTIC_V3_MAX_ATTEMPTS + 1,
    ):

        (
            system_message,
            user_message,
        ) = build_semantic_v3_prompt(
            batch_items,
            retry_feedback=(
                last_error
            ),
        )


        try:

            full_semantic_rate_limiter.wait_for_slot()


            api_calls += 1


            content = (
                call_semantic_v3_judge(
                    system_message,
                    user_message,
                )
            )


            model_payload = (
                parse_model_json(
                    content
                )
            )


            validated_judgments = (
                validate_semantic_v3_payload(
                    batch_items,
                    model_payload,
                )
            )


            break


        except Exception as exc:

            last_error = (
                f"{type(exc).__name__}: "
                f"{exc}"
            )


            attempt_errors.append({
                "attempt": (
                    attempt
                ),

                "error": (
                    last_error
                ),
            })


            normalized_error = (
                last_error.casefold()
            )


            is_rate_or_transport_error = any(
                token in normalized_error
                for token in [
                    "429",
                    "rate limit",
                    "too many requests",
                    "timeout",
                    "timed out",
                    "connection",
                    "502",
                    "503",
                    "504",
                ]
            )


            if (
                attempt
                < SEMANTIC_V3_MAX_ATTEMPTS
                and is_rate_or_transport_error
            ):

                time.sleep(
                    min(
                        20.0,
                        4.0
                        * float(
                            attempt
                        ),
                    )
                )


    if (
        validated_judgments
        is None
    ):

        return {
            "status": (
                "PENDING_RETRY"
            ),

            "batch_id": (
                batch_id
            ),

            "raw_claim_ids": [
                item[
                    "raw_claim_id"
                ]
                for item in (
                    batch_items
                )
            ],

            "api_calls": (
                api_calls
            ),

            "attempt_errors": (
                attempt_errors
            ),

            "last_error": (
                last_error
            ),
        }


    cache_payload = {
        "cache_version": (
            FULL_SEMANTIC_CACHE_VERSION
        ),

        "validation_version": (
            FULL_SEMANTIC_VERSION
        ),

        "batch_id": (
            batch_id
        ),

        "semantic_prompt_version": (
            SEMANTIC_V3_PROMPT_VERSION
        ),

        "model": (
            pilot_chat_model
        ),

        "source_raw_claims_sha256": (
            EXPECTED_FULL_RAW_SHA256
        ),

        "source_evidence_sha256": (
            EXPECTED_FULL_EVIDENCE_SHA256
        ),

        "validation_contract_sha256": (
            validation_contract_sha256
        ),

        "raw_claim_ids": [
            item[
                "raw_claim_id"
            ]
            for item in (
                batch_items
            )
        ],

        "judgments": (
            validated_judgments
        ),
    }


    write_status = (
        write_full_semantic_cache(
            cache_path,
            cache_payload,
        )
    )


    return {
        "status": (
            "VALID"
        ),

        "batch_id": (
            batch_id
        ),

        "raw_claim_ids": [
            item[
                "raw_claim_id"
            ]
            for item in (
                batch_items
            )
        ],

        "api_calls": (
            api_calls
        ),

        "attempt_errors": (
            attempt_errors
        ),

        "write_status": (
            write_status
        ),
    }


# ---------------------------------------------------------
# Validate every previously written Semantic cache.
# ---------------------------------------------------------

validated_full_semantic_cache_count = 0


for batch_items in (
    full_semantic_batches
):

    batch_id = (
        semantic_v3_batch_id(
            batch_items
        )
    )


    cache_path = (
        full_semantic_cache_path_for_batch(
            batch_id
        )
    )


    if not cache_path.exists():
        continue


    cached_payload = load_json(
        cache_path
    )


    validate_full_semantic_cache(
        batch_items,
        cached_payload,
    )


    validated_full_semantic_cache_count += 1


pending_full_semantic_batches = [
    batch_items
    for batch_items in (
        full_semantic_batches
    )
    if not (
        full_semantic_cache_path_for_batch(
            semantic_v3_batch_id(
                batch_items
            )
        ).exists()
    )
]


# Run ALL remaining batches in this execution.
semantic_batches_this_run = (
    pending_full_semantic_batches
)


# ---------------------------------------------------------
# Full parallel execution.
# ---------------------------------------------------------

full_semantic_api_calls_this_run = 0

full_semantic_new_caches_this_run = 0

full_semantic_failures_this_run = []

full_semantic_batches_completed_this_run = 0

full_semantic_retry_events_this_run = 0


print(
    "=== NB12i FULL SEMANTIC VALIDATION V3 — "
    "PARALLEL RATE-LIMITED ==="
)

print()

print(
    "Semantic Prompt version:",
    SEMANTIC_V3_PROMPT_VERSION,
)

print(
    "Model:",
    pilot_chat_model,
)

print(
    "Semantic Batch size:",
    SEMANTIC_V3_BATCH_SIZE,
)


print()

print(
    "Provider Rate Limit:",
    "30 requests/minute",
)

print(
    "Internal Rate Limit:",
    FULL_SEMANTIC_REQUESTS_PER_MINUTE,
    "requests/minute",
)

print(
    "Parallel Workers:",
    FULL_SEMANTIC_MAX_WORKERS,
)


print()

print(
    "Full Raw Claims:",
    len(
        full_semantic_items
    ),
)

print(
    "Pilot Semantic judgments reused:",
    len(
        pilot_semantic_reused_by_id
    ),
)

print(
    "Claims requiring Full Semantic API:",
    len(
        full_semantic_api_items
    ),
)

print(
    "Total Full Semantic API batches:",
    len(
        full_semantic_batches
    ),
)


print()

print(
    "Validated Semantic caches before run:",
    validated_full_semantic_cache_count,
)

print(
    "Pending Semantic batches before run:",
    len(
        pending_full_semantic_batches
    ),
)

print(
    "Semantic batches scheduled this run:",
    len(
        semantic_batches_this_run
    ),
)


if (
    semantic_batches_this_run
):

    theoretical_minutes = (
        len(
            semantic_batches_this_run
        )
        / float(
            FULL_SEMANTIC_REQUESTS_PER_MINUTE
        )
    )


    print(
        "Theoretical minimum request-start time:",
        round(
            theoretical_minutes,
            1,
        ),
        "minutes",
    )


print()


executor = ThreadPoolExecutor(
    max_workers=(
        FULL_SEMANTIC_MAX_WORKERS
    )
)


future_to_batch_id = {}


try:

    for batch_items in (
        semantic_batches_this_run
    ):

        batch_id = (
            semantic_v3_batch_id(
                batch_items
            )
        )


        future = executor.submit(
            process_full_semantic_batch,
            batch_items,
        )


        future_to_batch_id[
            future
        ] = (
            batch_id
        )


    for future in as_completed(
        future_to_batch_id
    ):

        result = future.result()


        full_semantic_batches_completed_this_run += 1


        full_semantic_api_calls_this_run += (
            result[
                "api_calls"
            ]
        )


        retry_count = len(
            result[
                "attempt_errors"
            ]
        )


        full_semantic_retry_events_this_run += (
            retry_count
        )


        if (
            result[
                "status"
            ]
            == "VALID"
        ):

            full_semantic_new_caches_this_run += 1

        else:

            full_semantic_failures_this_run.append(
                result
            )


        should_print_progress = (
            full_semantic_batches_completed_this_run
            % FULL_SEMANTIC_PROGRESS_EVERY
            == 0
            or result[
                "status"
            ]
            != "VALID"
            or retry_count
            > 0
            or full_semantic_batches_completed_this_run
            == len(
                semantic_batches_this_run
            )
        )


        if should_print_progress:

            print(
                "[",
                full_semantic_batches_completed_this_run,
                "/",
                len(
                    semantic_batches_this_run
                ),
                "]",
                result[
                    "status"
                ],
                "| batch:",
                result[
                    "batch_id"
                ],
                "| API calls:",
                result[
                    "api_calls"
                ],
                "| retries:",
                retry_count,
            )


            if retry_count:

                for error_item in (
                    result[
                        "attempt_errors"
                    ]
                ):

                    print(
                        "   attempt",
                        error_item[
                            "attempt"
                        ],
                        "INVALID |",
                        error_item[
                            "error"
                        ],
                    )


except KeyboardInterrupt:

    print()
    print(
        "KeyboardInterrupt received."
    )

    print(
        "Cancelling batches that have "
        "not started yet..."
    )


    for future in (
        future_to_batch_id
    ):

        future.cancel()


    executor.shutdown(
        wait=True,
        cancel_futures=True,
    )


    print(
        "Running workers finished."
    )

    print(
        "All already-written Semantic caches "
        "remain preserved."
    )


    raise


else:

    executor.shutdown(
        wait=True
    )


print()


# ---------------------------------------------------------
# Recount and independently revalidate every cache.
# ---------------------------------------------------------

validated_full_semantic_cache_count = 0

cached_full_semantic_judgments = []


for batch_items in (
    full_semantic_batches
):

    batch_id = (
        semantic_v3_batch_id(
            batch_items
        )
    )


    cache_path = (
        full_semantic_cache_path_for_batch(
            batch_id
        )
    )


    if not cache_path.exists():
        continue


    cached_payload = load_json(
        cache_path
    )


    validated_judgments = (
        validate_full_semantic_cache(
            batch_items,
            cached_payload,
        )
    )


    cached_full_semantic_judgments.extend(
        validated_judgments
    )


    validated_full_semantic_cache_count += 1


remaining_full_semantic_batches = (
    len(
        full_semantic_batches
    )
    - validated_full_semantic_cache_count
)


semantic_claims_completed = (
    len(
        pilot_semantic_reused_by_id
    )
    + len(
        cached_full_semantic_judgments
    )
)


print(
    "=== FULL SEMANTIC VALIDATION PROGRESS ==="
)

print()

print(
    "API calls this run:",
    full_semantic_api_calls_this_run,
)

print(
    "Retry events this run:",
    full_semantic_retry_events_this_run,
)

print(
    "New validated Semantic caches this run:",
    full_semantic_new_caches_this_run,
)


print()

print(
    "Validated Semantic caches:",
    validated_full_semantic_cache_count,
    "/",
    len(
        full_semantic_batches
    ),
)

print(
    "Claims with Semantic judgment:",
    semantic_claims_completed,
    "/",
    EXPECTED_FULL_RAW_CLAIMS,
)

print(
    "Remaining Semantic batches:",
    remaining_full_semantic_batches,
)

print(
    "Unresolved batches this run:",
    len(
        full_semantic_failures_this_run
    ),
)


if (
    full_semantic_failures_this_run
):

    print()

    print(
        "=== SEMANTIC BATCHES PENDING RETRY ==="
    )


    for failure in (
        full_semantic_failures_this_run
    ):

        print(
            "  -",
            failure[
                "batch_id"
            ],
            "| claims:",
            len(
                failure[
                    "raw_claim_ids"
                ]
            ),
            "|",
            failure[
                "last_error"
            ],
        )


# ---------------------------------------------------------
# Intermediate state.
# ---------------------------------------------------------

if (
    remaining_full_semantic_batches
    > 0
):

    assert not (
        full_semantic_adjudication_path.exists()
    ), (
        "Final Full Semantic adjudication "
        "already exists while caches "
        "are incomplete."
    )


    assert not (
        full_semantic_report_path.exists()
    ), (
        "Final Full Semantic report "
        "already exists while caches "
        "are incomplete."
    )


    full_semantic_validation_status = (
        "IN_PROGRESS"
    )


    print()

    print(
        "Successful Semantic caches preserved:",
        True,
    )

    print(
        "Final Semantic artifact published:",
        False,
    )

    print(
        "Automatic REPAIR publication:",
        False,
    )

    print(
        "Critical-risk final disposition:",
        "NOT_STARTED",
    )


    print()

    print(
        "CODE CELL 18 GATE:",
        full_semantic_validation_status,
    )


# ---------------------------------------------------------
# Complete state.
# ---------------------------------------------------------

else:

    assert not (
        full_semantic_failures_this_run
    ), (
        "All Semantic batches appear cached "
        "but this run still reports failures."
    )


    full_semantic_judgments = (
        list(
            pilot_semantic_reused_by_id.values()
        )
        + cached_full_semantic_judgments
    )


    full_semantic_judgments = sorted(
        full_semantic_judgments,
        key=lambda item: (
            item[
                "raw_claim_id"
            ]
        ),
    )


    full_semantic_ids = [
        item[
            "raw_claim_id"
        ]
        for item in (
            full_semantic_judgments
        )
    ]


    assert (
        len(
            full_semantic_ids
        )
        == EXPECTED_FULL_RAW_CLAIMS
    )


    assert (
        len(
            full_semantic_ids
        )
        == len(
            set(
                full_semantic_ids
            )
        )
    ), (
        "Duplicate Raw Claim IDs detected "
        "in Full Semantic judgments."
    )


    assert (
        set(
            full_semantic_ids
        )
        == set(
            full_semantic_claim_by_id
        )
    ), (
        "Full Semantic judgment coverage "
        "does not exactly match Raw Registry."
    )


    full_semantic_decision_counts = Counter(
        item[
            "decision"
        ]
        for item in (
            full_semantic_judgments
        )
    )


    full_semantic_issue_counts = Counter()

    full_semantic_failed_check_counts = Counter()

    full_semantic_source_mode_counts = Counter()

    full_semantic_decision_by_source = defaultdict(
        Counter
    )

    full_semantic_decision_by_role = defaultdict(
        Counter
    )


    for judgment in (
        full_semantic_judgments
    ):

        raw_claim_id = (
            judgment[
                "raw_claim_id"
            ]
        )


        full_semantic_source_mode_counts[
            judgment[
                "source_statement_mode"
            ]
        ] += 1


        for issue in (
            judgment[
                "issues"
            ]
        ):

            full_semantic_issue_counts[
                issue
            ] += 1


        for (
            check_name,
            check_value,
        ) in (
            judgment[
                "checks"
            ].items()
        ):

            if (
                check_value
                == "FAIL"
            ):

                full_semantic_failed_check_counts[
                    check_name
                ] += 1


        claim = (
            full_semantic_claim_by_id[
                raw_claim_id
            ]
        )


        evidence_items = (
            full_semantic_evidence_by_claim[
                raw_claim_id
            ]
        )


        source_types = {
            item[
                "source_type"
            ]
            for item in (
                evidence_items
            )
        }


        assert (
            len(
                source_types
            )
            == 1
        ), (
            "One Raw Claim unexpectedly "
            "contains multiple source types: "
            f"{raw_claim_id}"
        )


        source_type = next(
            iter(
                source_types
            )
        )


        decision = (
            judgment[
                "decision"
            ]
        )


        full_semantic_decision_by_source[
            source_type
        ][
            decision
        ] += 1


        full_semantic_decision_by_role[
            claim[
                "claim_discovery_role"
            ]
        ][
            decision
        ] += 1


    full_semantic_text = "".join(
        json.dumps(
            item,
            ensure_ascii=False,
            sort_keys=True,
        )
        + "\n"
        for item in (
            full_semantic_judgments
        )
    )


    full_semantic_sha256 = (
        hashlib.sha256(
            full_semantic_text.encode(
                "utf-8"
            )
        ).hexdigest()
    )


    full_semantic_report = {
        "report_version": (
            "claim_validation_full_semantic_report_v1"
        ),

        "validation_version": (
            FULL_SEMANTIC_VERSION
        ),

        "source_raw_claims": (
            full_raw_path.name
        ),

        "source_raw_claims_sha256": (
            EXPECTED_FULL_RAW_SHA256
        ),

        "source_evidence": (
            full_evidence_path.name
        ),

        "source_evidence_sha256": (
            EXPECTED_FULL_EVIDENCE_SHA256
        ),

        "validation_contract_version": (
            FINAL_VALIDATION_CONTRACT_VERSION
        ),

        "validation_contract_sha256": (
            validation_contract_sha256
        ),

        "semantic_prompt_version": (
            SEMANTIC_V3_PROMPT_VERSION
        ),

        "semantic_model": (
            pilot_chat_model
        ),

        "raw_claims": (
            EXPECTED_FULL_RAW_CLAIMS
        ),

        "claims_adjudicated": (
            len(
                full_semantic_judgments
            )
        ),

        "pilot_judgments_reused": (
            len(
                pilot_semantic_reused_by_id
            )
        ),

        "api_claims_adjudicated": (
            len(
                cached_full_semantic_judgments
            )
        ),

        "api_batches": (
            len(
                full_semantic_batches
            )
        ),

        "decision_counts": dict(
            sorted(
                full_semantic_decision_counts.items()
            )
        ),

        "failed_checks": dict(
            sorted(
                full_semantic_failed_check_counts.items()
            )
        ),

        "issue_counts": dict(
            sorted(
                full_semantic_issue_counts.items()
            )
        ),

        "source_statement_modes": dict(
            sorted(
                full_semantic_source_mode_counts.items()
            )
        ),

        "decision_counts_by_source_type": {
            source_type: dict(
                sorted(
                    counts.items()
                )
            )
            for (
                source_type,
                counts,
            ) in sorted(
                full_semantic_decision_by_source.items()
            )
        },

        "decision_counts_by_claim_discovery_role": {
            role: dict(
                sorted(
                    counts.items()
                )
            )
            for (
                role,
                counts,
            ) in sorted(
                full_semantic_decision_by_role.items()
            )
        },

        "duplicate_raw_claim_ids": 0,

        "missing_semantic_judgments": 0,

        "unresolved_batch_failures": 0,

        "automatic_repairs_published": 0,

        "critical_risk_final_disposition": (
            "NOT_STARTED"
        ),

        "validated_registry": (
            "NOT_YET_CREATED"
        ),

        "artifact": {
            "semantic_adjudication": (
                full_semantic_adjudication_path.name
            ),

            "semantic_adjudication_sha256": (
                full_semantic_sha256
            ),
        },
    }


    full_semantic_report_text = (
        json.dumps(
            full_semantic_report,
            ensure_ascii=False,
            indent=2,
            sort_keys=True,
        )
        + "\n"
    )


    full_semantic_write_status = (
        publish_frozen_artifact(
            full_semantic_adjudication_path,
            full_semantic_text,
        )
    )


    full_semantic_report_write_status = (
        publish_frozen_artifact(
            full_semantic_report_path,
            full_semantic_report_text,
        )
    )


    # -----------------------------------------------------
    # Independent disk revalidation.
    # -----------------------------------------------------

    assert (
        load_jsonl(
            full_semantic_adjudication_path
        )
        == full_semantic_judgments
    )


    assert (
        load_json(
            full_semantic_report_path
        )
        == full_semantic_report
    )


    assert (
        file_sha256(
            full_semantic_adjudication_path
        )
        == full_semantic_sha256
    )


    full_semantic_validation_status = (
        "PASS"
    )


    print()

    print(
        "=== NB12i FULL SEMANTIC VALIDATION RESULT ==="
    )


    print()

    print(
        "Claims adjudicated:",
        len(
            full_semantic_judgments
        ),
    )

    print(
        "Pilot judgments reused:",
        len(
            pilot_semantic_reused_by_id
        ),
    )

    print(
        "API-adjudicated Claims:",
        len(
            cached_full_semantic_judgments
        ),
    )


    print()

    print(
        "Semantic decisions:",
        dict(
            sorted(
                full_semantic_decision_counts.items()
            )
        ),
    )

    print(
        "Failed checks:",
        dict(
            sorted(
                full_semantic_failed_check_counts.items()
            )
        ),
    )

    print(
        "Issue counts:",
        dict(
            sorted(
                full_semantic_issue_counts.items()
            )
        ),
    )

    print(
        "Source statement modes:",
        dict(
            sorted(
                full_semantic_source_mode_counts.items()
            )
        ),
    )


    print()

    print(
        "Duplicate Raw Claim IDs:",
        0,
    )

    print(
        "Missing Semantic judgments:",
        0,
    )

    print(
        "Unresolved Semantic batches:",
        0,
    )


    print()

    print(
        "Semantic Adjudication Artifact:",
        full_semantic_adjudication_path.relative_to(
            project_root
        ),
    )

    print(
        "Semantic Adjudication SHA-256:",
        full_semantic_sha256,
    )

    print(
        "Write status:",
        full_semantic_write_status,
    )


    print()

    print(
        "Semantic Validation Report:",
        full_semantic_report_path.relative_to(
            project_root
        ),
    )

    print(
        "Report write status:",
        full_semantic_report_write_status,
    )


    print()

    print(
        "Automatic REPAIR publication:",
        False,
    )

    print(
        "Critical-risk final disposition:",
        "NOT_STARTED",
    )

    print(
        "Validated Registry:",
        "NOT_YET_CREATED",
    )


    print()

    print(
        "CODE CELL 18 GATE:",
        full_semantic_validation_status,
    )
```

<div style="direction: rtl; white-space: normal; line-height: 1.7;">

<h2>اعمال Final Disposition و ساخت Validated Claim Registry</h2>

<p>
اعتبارسنجی معنایی کامل روی تمام Claimهای خام با موفقیت پایان یافته است.
در این مرحله هیچ فراخوانی مدل انجام نمی‌شود و سیاست نهایی فریز‌شده Pilot
به‌صورت قطعی روی کل Registry اعمال می‌شود.
</p>

<p>
Claimهای دارای تصمیم
<code>REJECT</code>
رد می‌شوند،
Claimهای دارای تصمیم
<code>REPAIR</code>
برای بازبینی نگه داشته می‌شوند،
و Claimهای
<code>ACCEPT</code>
فقط زمانی وارد Registry معتبر می‌شوند که هیچ Critical Risk فریز‌شده‌ای نداشته باشند.
</p>

<ul>
<li>هیچ Semantic Repair به‌صورت خودکار منتشر نمی‌شود.</li>
<li>Claimهای ردشده یا Holdشده از Raw Registry حذف نمی‌شوند.</li>
<li>Evidence همه Claimها دوباره مستقیماً با Accepted Corpus تطبیق داده می‌شود.</li>
<li>خروجی معتبر در <code>claims_validated_v1.jsonl</code> منتشر می‌شود.</li>
<li>هنوز Deduplication انجام نمی‌شود و <code>claim_id</code> نهایی در این مرحله تخصیص داده نمی‌شود.</li>
<li>مرحله بعد فقط پس از PASS کامل این Gate، Deduplication خواهد بود.</li>
</ul>

</div>

```python
# Code Cell 19

assert (
    full_semantic_validation_status
    == "PASS"
), (
    "Code Cell 18 must PASS before "
    "Final Registry Validation."
)


FULL_VALIDATION_VERSION = (
    "claim_validation_full_v1"
)

FULL_DISPOSITION_VERSION = (
    "claim_validation_full_disposition_v1"
)


full_validation_disposition_path = (
    registry_artifact_root
    / "claim_validation_full_disposition_v1.jsonl"
)

validated_claims_path = (
    registry_artifact_root
    / "claims_validated_v1.jsonl"
)

full_validation_report_path = (
    registry_artifact_root
    / "claim_validation_full_report_v1.json"
)


# ---------------------------------------------------------
# Independently reload all authoritative inputs.
# ---------------------------------------------------------

validation_raw_claims = load_jsonl(
    full_raw_path
)

validation_evidence_links = load_jsonl(
    full_evidence_path
)

validation_semantic_judgments = load_jsonl(
    full_semantic_adjudication_path
)

validation_semantic_report = load_json(
    full_semantic_report_path
)

validation_frozen_contract = load_json(
    validation_contract_path
)


assert (
    validation_frozen_contract[
        "status"
    ]
    == "FROZEN_FOR_FULL_REGISTRY"
)


assert (
    validation_frozen_contract[
        "semantic_validation"
    ][
        "prompt_version"
    ]
    == SEMANTIC_V3_PROMPT_VERSION
)


assert (
    validation_frozen_contract[
        "semantic_validation"
    ][
        "automatic_repair_publish_allowed"
    ]
    is False
)


frozen_critical_risk_signals = set(
    validation_frozen_contract[
        "final_disposition_policy"
    ][
        "critical_risk_signals"
    ]
)


assert (
    frozen_critical_risk_signals
    == CRITICAL_RISK_SIGNALS
), (
    "Runtime Critical Risk policy differs "
    "from frozen Validation Contract."
)


# ---------------------------------------------------------
# Revalidate authoritative artifact identity.
# ---------------------------------------------------------

assert (
    len(
        validation_raw_claims
    )
    == EXPECTED_FULL_RAW_CLAIMS
)


assert (
    len(
        validation_evidence_links
    )
    == EXPECTED_FULL_EVIDENCE_LINKS
)


assert (
    len(
        validation_semantic_judgments
    )
    == EXPECTED_FULL_RAW_CLAIMS
)


assert (
    file_sha256(
        full_raw_path
    )
    == EXPECTED_FULL_RAW_SHA256
)


assert (
    file_sha256(
        full_evidence_path
    )
    == EXPECTED_FULL_EVIDENCE_SHA256
)


semantic_disk_sha256 = (
    file_sha256(
        full_semantic_adjudication_path
    )
)


assert (
    semantic_disk_sha256
    == validation_semantic_report[
        "artifact"
    ][
        "semantic_adjudication_sha256"
    ]
), (
    "Full Semantic adjudication SHA-256 "
    "does not match its report."
)


assert (
    validation_semantic_report[
        "claims_adjudicated"
    ]
    == EXPECTED_FULL_RAW_CLAIMS
)


assert (
    validation_semantic_report[
        "missing_semantic_judgments"
    ]
    == 0
)


assert (
    validation_semantic_report[
        "unresolved_batch_failures"
    ]
    == 0
)


# ---------------------------------------------------------
# Build authoritative indexes.
# ---------------------------------------------------------

validation_raw_by_id = {
    item[
        "raw_claim_id"
    ]: item
    for item in (
        validation_raw_claims
    )
}


validation_semantic_by_id = {
    item[
        "raw_claim_id"
    ]: item
    for item in (
        validation_semantic_judgments
    )
}


assert (
    len(
        validation_raw_by_id
    )
    == EXPECTED_FULL_RAW_CLAIMS
), (
    "Duplicate Raw Claim IDs detected."
)


assert (
    len(
        validation_semantic_by_id
    )
    == EXPECTED_FULL_RAW_CLAIMS
), (
    "Duplicate Semantic Judgment IDs detected."
)


assert (
    set(
        validation_raw_by_id
    )
    == set(
        validation_semantic_by_id
    )
), (
    "Semantic Judgment coverage does not "
    "exactly match Full Raw Claims."
)


# ---------------------------------------------------------
# Independently revalidate all Evidence against
# the frozen Accepted Corpus.
# ---------------------------------------------------------

validation_evidence_by_claim = defaultdict(
    list
)

validation_evidence_failures = []


for evidence in (
    validation_evidence_links
):

    raw_claim_id = evidence[
        "raw_claim_id"
    ]


    assert (
        raw_claim_id
        in validation_raw_by_id
    ), (
        "Evidence references unknown "
        f"Raw Claim: {raw_claim_id}"
    )


    document_id = evidence[
        "document_id"
    ]


    assert (
        document_id
        in accepted_records_by_id
    ), (
        "Evidence references unknown "
        f"document_id: {document_id}"
    )


    source_text = (
        accepted_records_by_id[
            document_id
        ][
            canonical_text_field
        ]
    )


    char_start = (
        evidence[
            "locator"
        ][
            "text_char_start"
        ]
    )

    char_end = (
        evidence[
            "locator"
        ][
            "text_char_end"
        ]
    )

    quote = (
        evidence[
            "evidence_quote_exact"
        ]
    )


    if (
        source_text[
            char_start:
            char_end
        ]
        != quote
    ):

        validation_evidence_failures.append(
            evidence[
                "evidence_id"
            ]
        )


    validation_evidence_by_claim[
        raw_claim_id
    ].append(
        evidence
    )


assert not (
    validation_evidence_failures
), (
    "Evidence no longer matches "
    "Accepted Corpus:\n"
    + "\n".join(
        validation_evidence_failures[
            :20
        ]
    )
)


for raw_claim_id in (
    validation_raw_by_id
):

    assert (
        validation_evidence_by_claim[
            raw_claim_id
        ]
    ), (
        "Raw Claim has no Evidence: "
        f"{raw_claim_id}"
    )


# ---------------------------------------------------------
# Apply the exact frozen Final Disposition policy.
#
# REJECT -> REJECT_SEMANTIC
# REPAIR -> HOLD_REPAIR_REQUIRED
# ACCEPT + Critical Risk -> HOLD_CRITICAL_RISK
# ACCEPT + no Critical Risk -> VALIDATED
#
# No API.
# No rewriting.
# No automatic repair publication.
# ---------------------------------------------------------

full_final_dispositions = []

full_validated_claims = []


for raw_claim_id in sorted(
    validation_raw_by_id
):

    claim = (
        validation_raw_by_id[
            raw_claim_id
        ]
    )

    judgment = (
        validation_semantic_by_id[
            raw_claim_id
        ]
    )


    semantic_decision = (
        judgment[
            "decision"
        ]
    )


    risk_signals = set(
        judgment.get(
            "fidelity_risk_signals",
            []
        )
    )


    triggered_critical_risks = sorted(
        risk_signals
        & frozen_critical_risk_signals
    )


    if (
        semantic_decision
        == "REJECT"
    ):

        final_disposition = (
            "REJECT_SEMANTIC"
        )


    elif (
        semantic_decision
        == "REPAIR"
    ):

        final_disposition = (
            "HOLD_REPAIR_REQUIRED"
        )


    else:

        assert (
            semantic_decision
            == "ACCEPT"
        )


        if triggered_critical_risks:

            final_disposition = (
                "HOLD_CRITICAL_RISK"
            )

        else:

            final_disposition = (
                "VALIDATED"
            )


    evidence_items = (
        validation_evidence_by_claim[
            raw_claim_id
        ]
    )


    source_types = sorted({
        item[
            "source_type"
        ]
        for item in (
            evidence_items
        )
    })


    assert (
        len(
            source_types
        )
        == 1
    ), (
        "One Raw Claim unexpectedly "
        "contains multiple source types: "
        f"{raw_claim_id}"
    )


    disposition_record = {
        "raw_claim_id": (
            raw_claim_id
        ),

        "final_disposition": (
            final_disposition
        ),

        "semantic_v3_decision": (
            semantic_decision
        ),

        "semantic_v3_source_statement_mode": (
            judgment[
                "source_statement_mode"
            ]
        ),

        "semantic_v3_issues": (
            judgment[
                "issues"
            ]
        ),

        "fidelity_risk_signals": sorted(
            risk_signals
        ),

        "triggered_critical_risk_signals": (
            triggered_critical_risks
        ),

        "claim_text_fa": (
            claim[
                "claim_text_fa"
            ]
        ),

        "claim_type": (
            claim[
                "claim_type"
            ]
        ),

        "source_type": (
            source_types[0]
        ),

        "authority_class": (
            claim[
                "authority_class"
            ]
        ),

        "claim_discovery_role": (
            claim[
                "claim_discovery_role"
            ]
        ),

        "evidence_ids": sorted(
            item[
                "evidence_id"
            ]
            for item in (
                evidence_items
            )
        ),

        "validation_contract_version": (
            FINAL_VALIDATION_CONTRACT_VERSION
        ),

        "validation_version": (
            FULL_VALIDATION_VERSION
        ),
    }


    full_final_dispositions.append(
        disposition_record
    )


    if (
        final_disposition
        == "VALIDATED"
    ):

        validated_claim = dict(
            claim
        )


        validated_claim[
            "registry_validation_status"
        ] = (
            "VALIDATED"
        )

        validated_claim[
            "validation_contract_version"
        ] = (
            FINAL_VALIDATION_CONTRACT_VERSION
        )

        validated_claim[
            "semantic_validation_version"
        ] = (
            SEMANTIC_V3_PROMPT_VERSION
        )

        validated_claim[
            "semantic_source_statement_mode"
        ] = (
            judgment[
                "source_statement_mode"
            ]
        )

        validated_claim[
            "semantic_issues"
        ] = []


        full_validated_claims.append(
            validated_claim
        )


# ---------------------------------------------------------
# Global Final Disposition validation.
# ---------------------------------------------------------

full_final_dispositions = sorted(
    full_final_dispositions,
    key=lambda item: (
        item[
            "raw_claim_id"
        ]
    ),
)


full_validated_claims = sorted(
    full_validated_claims,
    key=lambda item: (
        item[
            "raw_claim_id"
        ]
    ),
)


full_disposition_ids = [
    item[
        "raw_claim_id"
    ]
    for item in (
        full_final_dispositions
    )
]


assert (
    len(
        full_disposition_ids
    )
    == EXPECTED_FULL_RAW_CLAIMS
)


assert (
    len(
        full_disposition_ids
    )
    == len(
        set(
            full_disposition_ids
        )
    )
)


assert (
    set(
        full_disposition_ids
    )
    == set(
        validation_raw_by_id
    )
)


full_disposition_counts = Counter(
    item[
        "final_disposition"
    ]
    for item in (
        full_final_dispositions
    )
)


semantic_decision_counts_recomputed = Counter(
    item[
        "semantic_v3_decision"
    ]
    for item in (
        full_final_dispositions
    )
)


assert (
    dict(
        sorted(
            semantic_decision_counts_recomputed.items()
        )
    )
    == validation_semantic_report[
        "decision_counts"
    ]
), (
    "Semantic decision counts changed "
    "during Final Disposition."
)


# No Semantic REPAIR may be auto-published.
assert not any(
    item[
        "final_disposition"
    ]
    == "VALIDATED"
    and item[
        "semantic_v3_decision"
    ]
    == "REPAIR"
    for item in (
        full_final_dispositions
    )
)


# No Critical Risk may enter validated output.
assert not any(
    item[
        "final_disposition"
    ]
    == "VALIDATED"
    and item[
        "triggered_critical_risk_signals"
    ]
    for item in (
        full_final_dispositions
    )
)


critical_risk_accepts = [
    item
    for item in (
        full_final_dispositions
    )
    if (
        item[
            "semantic_v3_decision"
        ]
        == "ACCEPT"
        and item[
            "triggered_critical_risk_signals"
        ]
    )
]


assert all(
    item[
        "final_disposition"
    ]
    == "HOLD_CRITICAL_RISK"
    for item in (
        critical_risk_accepts
    )
)


# ---------------------------------------------------------
# Validate final VALIDATED Claim population.
# ---------------------------------------------------------

validated_claim_ids = {
    item[
        "raw_claim_id"
    ]
    for item in (
        full_validated_claims
    )
}


assert (
    len(
        validated_claim_ids
    )
    == len(
        full_validated_claims
    )
)


assert (
    len(
        validated_claim_ids
    )
    == full_disposition_counts[
        "VALIDATED"
    ]
)


validated_counts_by_source = Counter()

validated_counts_by_role = Counter()

validated_counts_by_type = Counter()


for claim in (
    full_validated_claims
):

    raw_claim_id = (
        claim[
            "raw_claim_id"
        ]
    )


    evidence_items = (
        validation_evidence_by_claim[
            raw_claim_id
        ]
    )


    source_types = {
        item[
            "source_type"
        ]
        for item in (
            evidence_items
        )
    }


    assert (
        len(
            source_types
        )
        == 1
    )


    source_type = next(
        iter(
            source_types
        )
    )


    validated_counts_by_source[
        source_type
    ] += 1


    validated_counts_by_role[
        claim[
            "claim_discovery_role"
        ]
    ] += 1


    validated_counts_by_type[
        claim[
            "claim_type"
        ]
    ] += 1


for source_type in [
    "book",
    "transcript",
    "website",
]:

    assert (
        validated_counts_by_source[
            source_type
        ]
        > 0
    ), (
        "Validated Registry lost "
        f"all {source_type} Claims."
    )


# ---------------------------------------------------------
# Serialize immutable artifacts.
# ---------------------------------------------------------

full_disposition_text = "".join(
    json.dumps(
        item,
        ensure_ascii=False,
        sort_keys=True,
    )
    + "\n"
    for item in (
        full_final_dispositions
    )
)


validated_claims_text = "".join(
    json.dumps(
        item,
        ensure_ascii=False,
        sort_keys=True,
    )
    + "\n"
    for item in (
        full_validated_claims
    )
)


full_disposition_sha256 = (
    hashlib.sha256(
        full_disposition_text.encode(
            "utf-8"
        )
    ).hexdigest()
)


validated_claims_sha256 = (
    hashlib.sha256(
        validated_claims_text.encode(
            "utf-8"
        )
    ).hexdigest()
)


full_validation_report = {
    "report_version": (
        "claim_validation_full_report_v1"
    ),

    "validation_version": (
        FULL_VALIDATION_VERSION
    ),

    "validation_contract_version": (
        FINAL_VALIDATION_CONTRACT_VERSION
    ),

    "validation_contract_sha256": (
        validation_contract_sha256
    ),

    "raw_claims": (
        len(
            validation_raw_claims
        )
    ),

    "semantic_judgments": (
        len(
            validation_semantic_judgments
        )
    ),

    "semantic_adjudication_sha256": (
        semantic_disk_sha256
    ),

    "semantic_decisions": dict(
        sorted(
            semantic_decision_counts_recomputed.items()
        )
    ),

    "final_dispositions": dict(
        sorted(
            full_disposition_counts.items()
        )
    ),

    "critical_risk_accepts_held": (
        len(
            critical_risk_accepts
        )
    ),

    "validated_claims": (
        len(
            full_validated_claims
        )
    ),

    "validated_by_source_type": dict(
        sorted(
            validated_counts_by_source.items()
        )
    ),

    "validated_by_claim_discovery_role": dict(
        sorted(
            validated_counts_by_role.items()
        )
    ),

    "validated_by_claim_type": dict(
        sorted(
            validated_counts_by_type.items()
        )
    ),

    "automatic_repairs_published": 0,

    "evidence_revalidation_failures": 0,

    "rejected_raw_claims_deleted": False,

    "held_raw_claims_deleted": False,

    "claim_ids_assigned": False,

    "deduplication": (
        "NOT_STARTED"
    ),

    "next_stage": (
        "CLAIM_DEDUPLICATION"
    ),

    "artifacts": {
        "full_disposition": (
            full_validation_disposition_path.name
        ),

        "full_disposition_sha256": (
            full_disposition_sha256
        ),

        "validated_claims": (
            validated_claims_path.name
        ),

        "validated_claims_sha256": (
            validated_claims_sha256
        ),
    },
}


full_validation_report_text = (
    json.dumps(
        full_validation_report,
        ensure_ascii=False,
        indent=2,
        sort_keys=True,
    )
    + "\n"
)


# ---------------------------------------------------------
# Immutable no-overwrite publication.
# ---------------------------------------------------------

full_disposition_write_status = (
    publish_frozen_artifact(
        full_validation_disposition_path,
        full_disposition_text,
    )
)


validated_claims_write_status = (
    publish_frozen_artifact(
        validated_claims_path,
        validated_claims_text,
    )
)


full_validation_report_write_status = (
    publish_frozen_artifact(
        full_validation_report_path,
        full_validation_report_text,
    )
)


# ---------------------------------------------------------
# Independent disk revalidation.
# ---------------------------------------------------------

assert (
    load_jsonl(
        full_validation_disposition_path
    )
    == full_final_dispositions
)


assert (
    load_jsonl(
        validated_claims_path
    )
    == full_validated_claims
)


assert (
    load_json(
        full_validation_report_path
    )
    == full_validation_report
)


assert (
    file_sha256(
        full_validation_disposition_path
    )
    == full_disposition_sha256
)


assert (
    file_sha256(
        validated_claims_path
    )
    == validated_claims_sha256
)


full_registry_validation_status = (
    "PASS"
)

full_registry_post_validation_readiness = (
    "READY_FOR_DEDUPLICATION"
)


# ---------------------------------------------------------
# Readable result.
# ---------------------------------------------------------

print(
    "=== NB12i FULL REGISTRY FINAL VALIDATION ==="
)

print()

print(
    "Raw Claims:",
    len(
        validation_raw_claims
    ),
)

print(
    "Semantic Decisions:",
    dict(
        sorted(
            semantic_decision_counts_recomputed.items()
        )
    ),
)


print()

print(
    "Final Dispositions:",
    dict(
        sorted(
            full_disposition_counts.items()
        )
    ),
)

print(
    "Critical-risk ACCEPTs moved to HOLD:",
    len(
        critical_risk_accepts
    ),
)

print(
    "Automatic REPAIR publications:",
    0,
)


print()

print(
    "Validated Claims:",
    len(
        full_validated_claims
    ),
)

print(
    "Validated by source type:",
    dict(
        sorted(
            validated_counts_by_source.items()
        )
    ),
)

print(
    "Validated by role:",
    dict(
        sorted(
            validated_counts_by_role.items()
        )
    ),
)

print(
    "Validated by Claim type:",
    dict(
        sorted(
            validated_counts_by_type.items()
        )
    ),
)


print()

print(
    "Evidence revalidation failures:",
    len(
        validation_evidence_failures
    ),
)

print(
    "Claim IDs assigned:",
    False,
)

print(
    "Deduplication:",
    "NOT_STARTED",
)


print()

print(
    "Validated Claims Artifact:",
    validated_claims_path.relative_to(
        project_root
    ),
)

print(
    "Validated Claims SHA-256:",
    validated_claims_sha256,
)

print(
    "Write status:",
    validated_claims_write_status,
)


print()

print(
    "Final Disposition Artifact:",
    full_validation_disposition_path.relative_to(
        project_root
    ),
)

print(
    "Final Disposition SHA-256:",
    full_disposition_sha256,
)

print(
    "Write status:",
    full_disposition_write_status,
)


print()

print(
    "Full Validation Report:",
    full_validation_report_path.relative_to(
        project_root
    ),
)

print(
    "Write status:",
    full_validation_report_write_status,
)


print()

print(
    "API call performed:",
    False,
)

print(
    "Next stage:",
    "CLAIM_DEDUPLICATION",
)

print(
    "Registry readiness:",
    full_registry_post_validation_readiness,
)


print()

print(
    "CODE CELL 19 GATE:",
    full_registry_validation_status,
)
```

<div style="direction: rtl; white-space: normal; line-height: 1.7;">

<h2>Deduplication مرحله اول: کشف Exact Duplicateها</h2>

<p>
پس از اعتبارسنجی معنایی و اعمال Critical-Risk Guard، تعداد ۴۰۳۹ Claim معتبر وارد مرحله Deduplication می‌شوند.
در این مرحله فقط تکرارهای کاملاً یکسان در متن Claim به‌صورت قطعی شناسایی می‌شوند.
</p>

<p>
این مرحله هیچ Semantic Merge، Canonicalization یا حذف Claim انجام نمی‌دهد.
تمام Claimها، Sourceها و Evidenceها حفظ می‌شوند و Exact Duplicate فقط به‌عنوان Relation ثبت می‌شود.
</p>

<ul>
<li>ورودی اصلی: <code>claims_validated_v1.jsonl</code></li>
<li>مقایسه Exact بر اساس متن ذخیره‌شده Claim انجام می‌شود.</li>
<li>گروه‌های Exact Duplicate و Pair Relationهای قطعی ساخته می‌شوند.</li>
<li>Claimهای منفرد بدون تغییر باقی می‌مانند.</li>
<li>هیچ API Call یا مدل زبانی استفاده نمی‌شود.</li>
<li>هیچ <code>claim_id</code> نهایی در این مرحله ساخته نمی‌شود.</li>
<li>خروجی این مرحله ورودی Candidate Generation برای Semantic Dedup خواهد بود.</li>
</ul>

</div>

```python
# Code Cell 20

assert (
    full_registry_validation_status
    == "PASS"
), (
    "Code Cell 19 must PASS before "
    "Claim Deduplication."
)

assert (
    full_registry_post_validation_readiness
    == "READY_FOR_DEDUPLICATION"
), (
    "Validated Registry is not ready "
    "for Deduplication."
)


DEDUP_EXACT_VERSION = (
    "claim_dedup_exact_v1"
)

DEDUP_EXACT_RELATION_TYPE = (
    "EXACT_DUPLICATE"
)


dedup_exact_groups_path = (
    registry_artifact_root
    / "claim_dedup_exact_groups_v1.jsonl"
)

dedup_exact_relations_path = (
    registry_artifact_root
    / "claim_dedup_exact_relations_v1.jsonl"
)

dedup_stage1_report_path = (
    registry_artifact_root
    / "claim_dedup_stage1_report_v1.json"
)


# ---------------------------------------------------------
# Independent reload of validated Registry.
# ---------------------------------------------------------

dedup_validated_claims = load_jsonl(
    validated_claims_path
)

dedup_full_evidence_links = load_jsonl(
    full_evidence_path
)

dedup_validation_report = load_json(
    full_validation_report_path
)

dedup_registry_contract = load_json(
    registry_contract_path
)


assert (
    file_sha256(
        validated_claims_path
    )
    == validated_claims_sha256
), (
    "Validated Claim Registry SHA-256 changed."
)


assert (
    len(
        dedup_validated_claims
    )
    == dedup_validation_report[
        "validated_claims"
    ]
), (
    "Validated Claim count differs "
    "from Final Validation Report."
)


assert (
    DEDUP_EXACT_RELATION_TYPE
    in dedup_registry_contract[
        "dedup_relation_types"
    ]
), (
    "EXACT_DUPLICATE is not present "
    "in the frozen Registry Contract."
)


assert (
    dedup_registry_contract[
        "relation_policy"
    ][
        "automatic_semantic_merge_allowed"
    ]
    is False
)


assert (
    dedup_registry_contract[
        "relation_policy"
    ][
        "automatic_subsumption_merge_allowed"
    ]
    is False
)


# ---------------------------------------------------------
# Validate Claim identities.
# ---------------------------------------------------------

dedup_claim_by_id = {
    claim[
        "raw_claim_id"
    ]: claim
    for claim in (
        dedup_validated_claims
    )
}


assert (
    len(
        dedup_claim_by_id
    )
    == len(
        dedup_validated_claims
    )
), (
    "Duplicate raw_claim_id detected "
    "in Validated Registry."
)


dedup_validated_ids = set(
    dedup_claim_by_id
)


# ---------------------------------------------------------
# Rebuild Evidence index for validated Claims only.
# ---------------------------------------------------------

dedup_evidence_by_claim = defaultdict(
    list
)


for evidence in (
    dedup_full_evidence_links
):

    raw_claim_id = evidence[
        "raw_claim_id"
    ]


    if (
        raw_claim_id
        not in dedup_validated_ids
    ):
        continue


    dedup_evidence_by_claim[
        raw_claim_id
    ].append(
        evidence
    )


for raw_claim_id in sorted(
    dedup_validated_ids
):

    assert (
        dedup_evidence_by_claim[
            raw_claim_id
        ]
    ), (
        "Validated Claim has no Evidence: "
        f"{raw_claim_id}"
    )


# ---------------------------------------------------------
# Exact-text grouping.
#
# IMPORTANT:
# This is intentionally strict.
#
# No normalization, paraphrase matching or semantic
# inference is applied here.
#
# Unicode / wording variants remain separate and will
# be handled only by later candidate generation.
# ---------------------------------------------------------

exact_claim_ids_by_text = defaultdict(
    list
)


for claim in (
    dedup_validated_claims
):

    claim_text = claim[
        "claim_text_fa"
    ]


    assert isinstance(
        claim_text,
        str,
    )


    assert (
        claim_text
    ), (
        "Validated Claim has empty text."
    )


    exact_claim_ids_by_text[
        claim_text
    ].append(
        claim[
            "raw_claim_id"
        ]
    )


exact_duplicate_text_groups = [
    (
        claim_text,
        sorted(
            raw_claim_ids
        ),
    )
    for (
        claim_text,
        raw_claim_ids,
    ) in (
        exact_claim_ids_by_text.items()
    )
    if (
        len(
            raw_claim_ids
        )
        >= 2
    )
]


exact_duplicate_text_groups = sorted(
    exact_duplicate_text_groups,
    key=lambda item: (
        item[0],
        item[1],
    ),
)


# ---------------------------------------------------------
# Build deterministic Exact Duplicate group records.
# ---------------------------------------------------------

dedup_exact_groups = []


for (
    claim_text,
    raw_claim_ids,
) in (
    exact_duplicate_text_groups
):

    text_sha256 = hashlib.sha256(
        claim_text.encode(
            "utf-8"
        )
    ).hexdigest()


    group_id = (
        "exact-claim-group:"
        + text_sha256[
            :24
        ]
    )


    group_source_types = set()

    group_authority_classes = set()

    group_roles = set()

    group_claim_types = set()

    group_document_ids = set()

    group_evidence_ids = set()


    for raw_claim_id in (
        raw_claim_ids
    ):

        claim = (
            dedup_claim_by_id[
                raw_claim_id
            ]
        )


        group_authority_classes.add(
            claim[
                "authority_class"
            ]
        )


        group_roles.add(
            claim[
                "claim_discovery_role"
            ]
        )


        group_claim_types.add(
            claim[
                "claim_type"
            ]
        )


        for document_id in (
            claim[
                "source_document_ids"
            ]
        ):

            group_document_ids.add(
                document_id
            )


        for evidence in (
            dedup_evidence_by_claim[
                raw_claim_id
            ]
        ):

            group_source_types.add(
                evidence[
                    "source_type"
                ]
            )


            group_evidence_ids.add(
                evidence[
                    "evidence_id"
                ]
            )


    group_record = {
        "exact_group_id": (
            group_id
        ),

        "relation_type": (
            DEDUP_EXACT_RELATION_TYPE
        ),

        "claim_text_fa": (
            claim_text
        ),

        "claim_text_sha256": (
            text_sha256
        ),

        "member_count": (
            len(
                raw_claim_ids
            )
        ),

        "raw_claim_ids": (
            raw_claim_ids
        ),

        "source_types": sorted(
            group_source_types
        ),

        "source_document_ids": sorted(
            group_document_ids
        ),

        "evidence_ids": sorted(
            group_evidence_ids
        ),

        "authority_classes": sorted(
            group_authority_classes
        ),

        "claim_discovery_roles": sorted(
            group_roles
        ),

        "claim_types": sorted(
            group_claim_types
        ),

        "cross_source_group": (
            len(
                group_source_types
            )
            > 1
        ),

        "automatic_merge_applied": False,

        "canonical_claim_selected": False,

        "dedup_version": (
            DEDUP_EXACT_VERSION
        ),
    }


    dedup_exact_groups.append(
        group_record
    )


# ---------------------------------------------------------
# Build deterministic pair relations.
#
# These relations are symmetric.
# Pair IDs are sorted before hashing.
#
# No Claim is deleted or merged.
# ---------------------------------------------------------

dedup_exact_relations = []


for group in (
    dedup_exact_groups
):

    member_ids = (
        group[
            "raw_claim_ids"
        ]
    )


    for left_index in range(
        len(
            member_ids
        )
    ):

        for right_index in range(
            left_index + 1,
            len(
                member_ids
            ),
        ):

            raw_claim_id_a = (
                member_ids[
                    left_index
                ]
            )

            raw_claim_id_b = (
                member_ids[
                    right_index
                ]
            )


            pair_key = (
                DEDUP_EXACT_RELATION_TYPE
                + "\n"
                + raw_claim_id_a
                + "\n"
                + raw_claim_id_b
            )


            relation_hash = hashlib.sha256(
                pair_key.encode(
                    "utf-8"
                )
            ).hexdigest()


            relation_id = (
                "claim-dedup-relation:"
                + relation_hash[
                    :24
                ]
            )


            source_types_a = sorted({
                evidence[
                    "source_type"
                ]
                for evidence in (
                    dedup_evidence_by_claim[
                        raw_claim_id_a
                    ]
                )
            })


            source_types_b = sorted({
                evidence[
                    "source_type"
                ]
                for evidence in (
                    dedup_evidence_by_claim[
                        raw_claim_id_b
                    ]
                )
            })


            relation_record = {
                "relation_id": (
                    relation_id
                ),

                "relation_type": (
                    DEDUP_EXACT_RELATION_TYPE
                ),

                "raw_claim_id_a": (
                    raw_claim_id_a
                ),

                "raw_claim_id_b": (
                    raw_claim_id_b
                ),

                "exact_group_id": (
                    group[
                        "exact_group_id"
                    ]
                ),

                "claim_text_sha256": (
                    group[
                        "claim_text_sha256"
                    ]
                ),

                "source_types_a": (
                    source_types_a
                ),

                "source_types_b": (
                    source_types_b
                ),

                "cross_source_pair": (
                    set(
                        source_types_a
                    )
                    != set(
                        source_types_b
                    )
                ),

                "decision_basis": (
                    "IDENTICAL_STORED_CLAIM_TEXT"
                ),

                "automatic_merge_applied": False,

                "dedup_version": (
                    DEDUP_EXACT_VERSION
                ),
            }


            dedup_exact_relations.append(
                relation_record
            )


dedup_exact_relations = sorted(
    dedup_exact_relations,
    key=lambda item: (
        item[
            "relation_id"
        ]
    ),
)


# ---------------------------------------------------------
# Global deterministic validation.
# ---------------------------------------------------------

exact_group_ids = [
    item[
        "exact_group_id"
    ]
    for item in (
        dedup_exact_groups
    )
]


exact_relation_ids = [
    item[
        "relation_id"
    ]
    for item in (
        dedup_exact_relations
    )
]


assert (
    len(
        exact_group_ids
    )
    == len(
        set(
            exact_group_ids
        )
    )
)


assert (
    len(
        exact_relation_ids
    )
    == len(
        set(
            exact_relation_ids
        )
    )
)


claims_in_exact_groups = {
    raw_claim_id
    for group in (
        dedup_exact_groups
    )
    for raw_claim_id in (
        group[
            "raw_claim_ids"
        ]
    )
}


exact_group_member_count = sum(
    group[
        "member_count"
    ]
    for group in (
        dedup_exact_groups
    )
)


assert (
    exact_group_member_count
    >= len(
        claims_in_exact_groups
    )
)


exact_singleton_claim_count = (
    len(
        dedup_validated_claims
    )
    - len(
        claims_in_exact_groups
    )
)


exact_unique_text_count = (
    len(
        exact_claim_ids_by_text
    )
)


assert (
    exact_unique_text_count
    <= len(
        dedup_validated_claims
    )
)


cross_source_exact_groups = [
    item
    for item in (
        dedup_exact_groups
    )
    if item[
        "cross_source_group"
    ]
]


cross_source_exact_relations = [
    item
    for item in (
        dedup_exact_relations
    )
    if item[
        "cross_source_pair"
    ]
]


all_pair_comparison_count = (
    len(
        dedup_validated_claims
    )
    * (
        len(
            dedup_validated_claims
        )
        - 1
    )
    // 2
)


# ---------------------------------------------------------
# Serialize intermediate Dedup artifacts.
# ---------------------------------------------------------

dedup_exact_groups_text = "".join(
    json.dumps(
        item,
        ensure_ascii=False,
        sort_keys=True,
    )
    + "\n"
    for item in (
        dedup_exact_groups
    )
)


dedup_exact_relations_text = "".join(
    json.dumps(
        item,
        ensure_ascii=False,
        sort_keys=True,
    )
    + "\n"
    for item in (
        dedup_exact_relations
    )
)


dedup_exact_groups_sha256 = (
    hashlib.sha256(
        dedup_exact_groups_text.encode(
            "utf-8"
        )
    ).hexdigest()
)


dedup_exact_relations_sha256 = (
    hashlib.sha256(
        dedup_exact_relations_text.encode(
            "utf-8"
        )
    ).hexdigest()
)


dedup_stage1_report = {
    "report_version": (
        "claim_dedup_stage1_report_v1"
    ),

    "dedup_version": (
        DEDUP_EXACT_VERSION
    ),

    "source_validated_claims": (
        validated_claims_path.name
    ),

    "source_validated_claims_sha256": (
        validated_claims_sha256
    ),

    "validated_claims": (
        len(
            dedup_validated_claims
        )
    ),

    "exact_unique_claim_texts": (
        exact_unique_text_count
    ),

    "exact_duplicate_groups": (
        len(
            dedup_exact_groups
        )
    ),

    "claims_in_exact_duplicate_groups": (
        len(
            claims_in_exact_groups
        )
    ),

    "exact_singleton_claims": (
        exact_singleton_claim_count
    ),

    "exact_duplicate_pair_relations": (
        len(
            dedup_exact_relations
        )
    ),

    "cross_source_exact_groups": (
        len(
            cross_source_exact_groups
        )
    ),

    "cross_source_exact_pair_relations": (
        len(
            cross_source_exact_relations
        )
    ),

    "theoretical_all_pairs_without_candidate_filter": (
        all_pair_comparison_count
    ),

    "normalization_applied": False,

    "semantic_similarity_used": False,

    "model_calls": 0,

    "automatic_merge_applied": False,

    "canonical_claim_selected": False,

    "claim_ids_assigned": False,

    "semantic_deduplication": (
        "NOT_STARTED"
    ),

    "next_stage": (
        "SEMANTIC_DEDUP_CANDIDATE_GENERATION"
    ),

    "artifacts": {
        "exact_groups": (
            dedup_exact_groups_path.name
        ),

        "exact_groups_sha256": (
            dedup_exact_groups_sha256
        ),

        "exact_relations": (
            dedup_exact_relations_path.name
        ),

        "exact_relations_sha256": (
            dedup_exact_relations_sha256
        ),
    },
}


dedup_stage1_report_text = (
    json.dumps(
        dedup_stage1_report,
        ensure_ascii=False,
        indent=2,
        sort_keys=True,
    )
    + "\n"
)


# ---------------------------------------------------------
# Immutable publication.
# ---------------------------------------------------------

dedup_exact_groups_write_status = (
    publish_frozen_artifact(
        dedup_exact_groups_path,
        dedup_exact_groups_text,
    )
)


dedup_exact_relations_write_status = (
    publish_frozen_artifact(
        dedup_exact_relations_path,
        dedup_exact_relations_text,
    )
)


dedup_stage1_report_write_status = (
    publish_frozen_artifact(
        dedup_stage1_report_path,
        dedup_stage1_report_text,
    )
)


# ---------------------------------------------------------
# Independent disk revalidation.
# ---------------------------------------------------------

assert (
    load_jsonl(
        dedup_exact_groups_path
    )
    == dedup_exact_groups
)


assert (
    load_jsonl(
        dedup_exact_relations_path
    )
    == dedup_exact_relations
)


assert (
    load_json(
        dedup_stage1_report_path
    )
    == dedup_stage1_report
)


assert (
    file_sha256(
        dedup_exact_groups_path
    )
    == dedup_exact_groups_sha256
)


assert (
    file_sha256(
        dedup_exact_relations_path
    )
    == dedup_exact_relations_sha256
)


claim_exact_dedup_status = "PASS"

claim_dedup_readiness = (
    "READY_FOR_SEMANTIC_CANDIDATE_GENERATION"
)


# ---------------------------------------------------------
# Result.
# ---------------------------------------------------------

print(
    "=== NB12i CLAIM DEDUP STAGE 1 — EXACT ==="
)

print()

print(
    "Validated Claims:",
    len(
        dedup_validated_claims
    ),
)

print(
    "Exact unique Claim texts:",
    exact_unique_text_count,
)

print(
    "Exact Duplicate groups:",
    len(
        dedup_exact_groups
    ),
)

print(
    "Claims inside Exact groups:",
    len(
        claims_in_exact_groups
    ),
)

print(
    "Exact singleton Claims:",
    exact_singleton_claim_count,
)

print(
    "Exact Duplicate pair relations:",
    len(
        dedup_exact_relations
    ),
)


print()

print(
    "Cross-source Exact groups:",
    len(
        cross_source_exact_groups
    ),
)

print(
    "Cross-source Exact pair relations:",
    len(
        cross_source_exact_relations
    ),
)


print()

print(
    "Theoretical all-pairs comparisons:",
    all_pair_comparison_count,
)

print(
    "Semantic similarity used:",
    False,
)

print(
    "API calls performed:",
    0,
)

print(
    "Automatic merge applied:",
    False,
)

print(
    "Claim IDs assigned:",
    False,
)


print()

print(
    "Exact Groups Artifact:",
    dedup_exact_groups_path.relative_to(
        project_root
    ),
)

print(
    "Exact Groups SHA-256:",
    dedup_exact_groups_sha256,
)

print(
    "Write status:",
    dedup_exact_groups_write_status,
)


print()

print(
    "Exact Relations Artifact:",
    dedup_exact_relations_path.relative_to(
        project_root
    ),
)

print(
    "Exact Relations SHA-256:",
    dedup_exact_relations_sha256,
)

print(
    "Write status:",
    dedup_exact_relations_write_status,
)


print()

print(
    "Next stage:",
    "SEMANTIC_DEDUP_CANDIDATE_GENERATION",
)

print(
    "Dedup readiness:",
    claim_dedup_readiness,
)


print()

print(
    "CODE CELL 20 GATE:",
    claim_exact_dedup_status,
)
```

<h2>Semantic Dedup Candidate Generation — Embedding Runtime Handoff</h2>

<p>
مرحله Exact Dedup با موفقیت پایان یافته است. اکنون برای جلوگیری از مقایسه همه‌به‌همه،
Candidateهای Semantic Dedup باید با Embedding ساخته شوند.
</p>

<p>
این Cell فقط تنظیمات واقعی Embedding موجود در پروژه را از Notebook مرجع شناسایی می‌کند.
هیچ درخواست API، تولید Embedding، تغییر Artifact یا هزینه‌ای در این مرحله انجام نمی‌شود.
</p>

<ul>
<li>تعداد Claimهای معتبر: ۴۰۳۹</li>
<li>تعداد متن‌های یکتای Exact: ۴۰۰۶</li>
<li>مقایسه نظری همه‌به‌همه: ۸٬۱۵۴٬۷۴۱ جفت</li>
<li>هدف مرحله بعد: ساخت Candidate Pool محدود با Embedding و سپس داوری فقط همان Candidateها</li>
</ul>

```python
# Code Cell 21

assert (
    claim_exact_dedup_status
    == "PASS"
), (
    "Code Cell 20 must PASS before "
    "Semantic Dedup Candidate Generation."
)

assert (
    claim_dedup_readiness
    == "READY_FOR_SEMANTIC_CANDIDATE_GENERATION"
)


# ---------------------------------------------------------
# Targeted read-only discovery of the project's real
# Embedding runtime.
#
# No API calls.
# No files written.
# No secrets printed.
# ---------------------------------------------------------

embedding_reference_notebook = (
    project_root
    / "notebooks"
    / "phase2"
    / "13_embedding_benchmark.ipynb"
)

assert (
    embedding_reference_notebook.exists()
), (
    "Embedding benchmark notebook not found: "
    f"{embedding_reference_notebook}"
)


embedding_nb = json.loads(
    embedding_reference_notebook.read_text(
        encoding="utf-8"
    )
)


embedding_code_cells = [
    cell
    for cell in embedding_nb.get(
        "cells",
        []
    )
    if (
        cell.get("cell_type")
        == "code"
    )
]


embedding_runtime_evidence = []

embedding_model_evidence = []

embedding_endpoint_evidence = []

embedding_request_evidence = []

embedding_env_names = set()


SENSITIVE_TOKENS = (
    "api_key",
    "apikey",
    "secret",
    "token",
    "authorization",
    "bearer",
)


for cell_index, cell in enumerate(
    embedding_code_cells,
    start=1,
):

    source = "".join(
        cell.get(
            "source",
            []
        )
    )


    for line_number, raw_line in enumerate(
        source.splitlines(),
        start=1,
    ):

        stripped = (
            raw_line.strip()
        )

        lowered = (
            stripped.casefold()
        )


        embedding_related = any(
            token in lowered
            for token in (
                "embedding",
                "embeddings.create",
                "text-embedding",
                "openai(",
                "base_url",
            )
        )


        if not embedding_related:
            continue


        secret_bearing = any(
            token in lowered
            for token in (
                "api_key",
                "apikey",
                "secret",
                "authorization",
                "bearer",
            )
        )


        safe_line = (
            "<REDACTED SECRET-BEARING LINE>"
            if secret_bearing
            else stripped
        )


        record = {
            "cell": (
                cell_index
            ),

            "line": (
                line_number
            ),

            "text": (
                safe_line
            ),
        }


        embedding_runtime_evidence.append(
            record
        )


        if any(
            token in lowered
            for token in (
                "embedding_model",
                "embedding-model",
                "text-embedding",
                '"model"',
                "model=",
            )
        ):

            embedding_model_evidence.append(
                record
            )


        if any(
            token in lowered
            for token in (
                "base_url",
                "embedding_url",
                "endpoint",
            )
        ):

            embedding_endpoint_evidence.append(
                record
            )


        if (
            "embeddings.create"
            in lowered
        ):

            embedding_request_evidence.append(
                record
            )


        # Discover environment variable names only.
        # Values are never printed.
        marker = 'getenv("'

        cursor = 0

        while True:

            marker_index = (
                raw_line.find(
                    marker,
                    cursor,
                )
            )


            if marker_index < 0:
                break


            name_start = (
                marker_index
                + len(marker)
            )

            name_end = (
                raw_line.find(
                    '"',
                    name_start,
                )
            )


            if name_end < 0:
                break


            env_name = (
                raw_line[
                    name_start:
                    name_end
                ]
            )


            if any(
                token in env_name.casefold()
                for token in (
                    "embed",
                    "arvan",
                    "openai",
                )
            ):

                embedding_env_names.add(
                    env_name
                )


            cursor = (
                name_end
                + 1
            )


# ---------------------------------------------------------
# Runtime globals already available in the current kernel.
# Values which may contain credentials are not printed.
# ---------------------------------------------------------

embedding_runtime_global_names = sorted(
    name
    for name in globals()
    if (
        "embed"
        in name.casefold()
    )
)


embedding_runtime_globals = []


for name in (
    embedding_runtime_global_names
):

    value = globals()[
        name
    ]


    if any(
        token in name.casefold()
        for token in SENSITIVE_TOKENS
    ):

        value_preview = (
            "<REDACTED>"
        )

    elif isinstance(
        value,
        (
            str,
            int,
            float,
            bool,
            type(None),
        ),
    ):

        value_preview = (
            value
        )

    else:

        value_preview = (
            f"<{type(value).__name__}>"
        )


    embedding_runtime_globals.append({
        "name": (
            name
        ),

        "type": (
            type(value).__name__
        ),

        "value_preview": (
            value_preview
        ),
    })


# ---------------------------------------------------------
# Verify the exact-dedup handoff population.
# ---------------------------------------------------------

embedding_handoff_claims = load_jsonl(
    validated_claims_path
)


embedding_unique_texts = {
    claim[
        "claim_text_fa"
    ]
    for claim in (
        embedding_handoff_claims
    )
}


assert (
    len(
        embedding_handoff_claims
    )
    == 4039
)


assert (
    len(
        embedding_unique_texts
    )
    == 4006
)


# ---------------------------------------------------------
# Gate.
# ---------------------------------------------------------

assert (
    embedding_request_evidence
), (
    "No embeddings.create usage found in "
    "Notebook 13."
)


embedding_runtime_handoff_status = (
    "PASS"
)


print(
    "=== NB12i EMBEDDING RUNTIME HANDOFF ==="
)

print()

print(
    "Validated Claims:",
    len(
        embedding_handoff_claims
    ),
)

print(
    "Exact-unique Claim texts:",
    len(
        embedding_unique_texts
    ),
)

print(
    "Theoretical all-pairs:",
    (
        len(
            embedding_handoff_claims
        )
        * (
            len(
                embedding_handoff_claims
            )
            - 1
        )
        // 2
    ),
)


print()

print(
    "Reference Notebook:",
    embedding_reference_notebook.relative_to(
        project_root
    ),
)

print(
    "Code cells scanned:",
    len(
        embedding_code_cells
    ),
)

print(
    "Embedding runtime evidence lines:",
    len(
        embedding_runtime_evidence
    ),
)

print(
    "Embedding model evidence lines:",
    len(
        embedding_model_evidence
    ),
)

print(
    "Embedding endpoint evidence lines:",
    len(
        embedding_endpoint_evidence
    ),
)

print(
    "embeddings.create evidence lines:",
    len(
        embedding_request_evidence
    ),
)


print()

print(
    "Embedding-related environment names:"
)

for name in sorted(
    embedding_env_names
):

    print(
        "  -",
        name,
    )


print()

print(
    "Current kernel embedding-related globals:"
)

if (
    embedding_runtime_globals
):

    for item in (
        embedding_runtime_globals
    ):

        print(
            "  -",
            item[
                "name"
            ],
            "| type:",
            item[
                "type"
            ],
            "| value:",
            item[
                "value_preview"
            ],
        )

else:

    print(
        "  - NONE"
    )


print()

print(
    "Relevant Notebook 13 evidence:"
)

for record in (
    embedding_runtime_evidence[
        :40
    ]
):

    print(
        "  - cell",
        record[
            "cell"
        ],
        "| line",
        record[
            "line"
        ],
        "|",
        record[
            "text"
        ],
    )


if (
    len(
        embedding_runtime_evidence
    )
    > 40
):

    print(
        "  ...",
        len(
            embedding_runtime_evidence
        )
        - 40,
        "more lines",
    )


print()

print(
    "API calls performed:",
    0,
)

print(
    "Files modified:",
    False,
)

print(
    "Secrets printed:",
    False,
)


print()

print(
    "CODE CELL 21 GATE:",
    embedding_runtime_handoff_status,
)
```

<h2>Semantic Dedup Candidate Generation با Embedding</h2>

<p>
پس از حذف Exact Duplicateها از فضای مقایسه، ۴۰۰۶ متن Claim یکتا باقی مانده است.
در این مرحله برای هر متن یک Embedding ساخته می‌شود و نزدیک‌ترین همسایه‌های معنایی آن
با Cosine Similarity محاسبه می‌شوند.
</p>

<p>
این مرحله فقط Candidate Retrieval است و هیچ تصمیمی درباره Duplicate بودن،
Subsumption یا Partial Overlap نمی‌گیرد.
به همین دلیل هنوز هیچ Semantic Merge یا LLM Adjudication انجام نمی‌شود.
</p>

<ul>
<li>Embedding Model: <code>text-embedding-3-large</code></li>
<li>Embedding فقط برای ۴۰۰۶ متن Exact-Unique ساخته می‌شود.</li>
<li>درخواست‌ها Batch و Cache می‌شوند تا اجرای مجدد هزینه تکراری نداشته باشد.</li>
<li>برای هر متن ۵۰ همسایه نزدیک نگه داشته می‌شود تا Recall بالا بماند.</li>
<li>تمام Similarityها به‌صورت محلی محاسبه می‌شوند.</li>
<li>Exact Duplicateهای مرحله قبل دوباره وارد Semantic Candidateها نمی‌شوند.</li>
<li>هیچ Claim حذف یا Merge نمی‌شود و هنوز <code>claim_id</code> نهایی ساخته نمی‌شود.</li>
</ul>

</div>

```python
# Code Cell 22

# Temporary development imports.
# Move these to Code Cell 01 before final Restart -> Run All.
import os
import time
import numpy as np
from openai import OpenAI


assert (
    embedding_runtime_handoff_status == "PASS"
), (
    "Code Cell 21 must PASS before "
    "Embedding Candidate Generation."
)

assert (
    claim_exact_dedup_status == "PASS"
)


# =========================================================
# Configuration
# =========================================================

EMBEDDING_STAGE_VERSION = (
    "claim_semantic_candidate_generation_v1"
)

EMBEDDING_CACHE_VERSION = (
    "claim_embedding_cache_v1"
)

EMBEDDING_BATCH_SIZE = 64
EMBEDDING_INTERNAL_RPM = 8
EMBEDDING_MAX_ATTEMPTS = 4

SEMANTIC_NEIGHBOR_TOP_K = 50
SIMILARITY_BLOCK_SIZE = 256


embedding_cache_root = (
    registry_artifact_root
    / "claim_embedding_cache_v1"
)

embedding_text_index_path = (
    registry_artifact_root
    / "claim_embedding_text_index_v1.jsonl"
)

embedding_matrix_path = (
    registry_artifact_root
    / "claim_embedding_matrix_v1.npz"
)

semantic_neighbor_candidates_path = (
    registry_artifact_root
    / "claim_semantic_neighbor_candidates_v1.jsonl"
)

semantic_candidate_report_path = (
    registry_artifact_root
    / "claim_semantic_candidate_generation_report_v1.json"
)


embedding_cache_root.mkdir(
    parents=True,
    exist_ok=True,
)


# =========================================================
# Existing frozen report, if this is a rerun
# =========================================================

existing_semantic_candidate_report = None

if semantic_candidate_report_path.exists():
    existing_semantic_candidate_report = load_json(
        semantic_candidate_report_path
    )


# =========================================================
# Resolve exact Embedding model from Notebook 13
# =========================================================

reference_sources = [
    "".join(
        cell.get("source", [])
    )
    for cell in embedding_code_cells
]

reference_lines = [
    line
    for source in reference_sources
    for line in source.splitlines()
]


resolved_embedding_model_id = None

for raw_line in reference_lines:

    stripped = raw_line.strip()

    if (
        stripped.startswith(
            "arvan_embedding_model_id"
        )
        and "=" in stripped
    ):

        rhs = (
            stripped
            .split("=", 1)[1]
            .strip()
        )

        if rhs.endswith(","):
            rhs = rhs[:-1].strip()

        try:
            candidate_value = json.loads(
                rhs
            )

        except Exception:
            continue

        if isinstance(
            candidate_value,
            str,
        ):
            resolved_embedding_model_id = (
                candidate_value
            )
            break


assert (
    resolved_embedding_model_id
), (
    "Could not resolve Embedding Model ID "
    "from Notebook 13."
)

assert (
    resolved_embedding_model_id
    == "text-embedding-3-large"
), (
    "Unexpected Embedding Model ID: "
    f"{resolved_embedding_model_id}"
)


# =========================================================
# Independently rebuild exact-unique Claim-text population
# =========================================================

candidate_validated_claims = load_jsonl(
    validated_claims_path
)

dedup_stage1_disk_report = load_json(
    dedup_stage1_report_path
)


assert (
    len(candidate_validated_claims)
    == dedup_stage1_disk_report[
        "validated_claims"
    ]
)


claim_ids_by_exact_text = defaultdict(
    list
)

for claim in candidate_validated_claims:

    claim_ids_by_exact_text[
        claim["claim_text_fa"]
    ].append(
        claim["raw_claim_id"]
    )


assert (
    len(claim_ids_by_exact_text)
    == dedup_stage1_disk_report[
        "exact_unique_claim_texts"
    ]
), (
    "Exact-unique Claim-text count differs "
    "from Code Cell 20 report."
)


embedding_text_records = []

for claim_text in sorted(
    claim_ids_by_exact_text
):

    text_sha256 = hashlib.sha256(
        claim_text.encode(
            "utf-8"
        )
    ).hexdigest()

    text_id = (
        "claim-text:"
        + text_sha256[:24]
    )

    member_ids = sorted(
        claim_ids_by_exact_text[
            claim_text
        ]
    )

    embedding_text_records.append({
        "text_id": text_id,
        "claim_text_sha256": (
            text_sha256
        ),
        "claim_text_fa": (
            claim_text
        ),
        "raw_claim_ids": (
            member_ids
        ),
        "member_count": (
            len(member_ids)
        ),
    })


embedding_text_records = sorted(
    embedding_text_records,
    key=lambda item: (
        item["text_id"]
    ),
)


embedding_text_ids = [
    item["text_id"]
    for item in embedding_text_records
]


assert (
    len(embedding_text_ids)
    == len(set(embedding_text_ids))
)


# =========================================================
# Freeze / reuse deterministic text index
# =========================================================

embedding_text_index_text = "".join(
    json.dumps(
        item,
        ensure_ascii=False,
        sort_keys=True,
    )
    + "\n"
    for item in embedding_text_records
)


embedding_text_index_sha256 = (
    hashlib.sha256(
        embedding_text_index_text.encode(
            "utf-8"
        )
    ).hexdigest()
)


embedding_text_index_write_status = (
    publish_frozen_artifact(
        embedding_text_index_path,
        embedding_text_index_text,
    )
)


assert (
    file_sha256(
        embedding_text_index_path
    )
    == embedding_text_index_sha256
)


# =========================================================
# Deterministic Embedding batches
# =========================================================

embedding_batches = [
    embedding_text_records[
        start:
        start + EMBEDDING_BATCH_SIZE
    ]
    for start in range(
        0,
        len(embedding_text_records),
        EMBEDDING_BATCH_SIZE,
    )
]


assert all(
    1 <= len(batch)
    <= EMBEDDING_BATCH_SIZE
    for batch in embedding_batches
)


def embedding_batch_id(
    batch_records,
):

    joined_ids = "\n".join(
        item["text_id"]
        for item in batch_records
    )

    digest = hashlib.sha256(
        joined_ids.encode(
            "utf-8"
        )
    ).hexdigest()

    return (
        "embedding-batch:"
        + digest[:24]
    )


def embedding_cache_path_for_batch(
    batch_records,
):

    batch_id = embedding_batch_id(
        batch_records
    )

    return (
        embedding_cache_root
        / (
            batch_id.replace(
                ":",
                "_",
            )
            + ".npz"
        )
    )


# =========================================================
# Existing Embedding-cache validator
# =========================================================

def load_and_validate_embedding_cache(
    batch_records,
    cache_path,
):

    with np.load(
        cache_path,
        allow_pickle=False,
    ) as payload:

        cached_version = str(
            payload[
                "cache_version"
            ].item()
        )

        cached_model = str(
            payload[
                "model_id"
            ].item()
        )

        cached_batch_id = str(
            payload[
                "batch_id"
            ].item()
        )

        cached_text_ids = [
            str(value)
            for value in payload[
                "text_ids"
            ].tolist()
        ]

        cached_vectors = np.asarray(
            payload["vectors"],
            dtype=np.float32,
        )


    expected_text_ids = [
        item["text_id"]
        for item in batch_records
    ]


    assert (
        cached_version
        == EMBEDDING_CACHE_VERSION
    )

    assert (
        cached_model
        == resolved_embedding_model_id
    )

    assert (
        cached_batch_id
        == embedding_batch_id(
            batch_records
        )
    )

    assert (
        cached_text_ids
        == expected_text_ids
    )

    assert (
        cached_vectors.ndim
        == 2
    )

    assert (
        cached_vectors.shape[0]
        == len(batch_records)
    )

    assert np.isfinite(
        cached_vectors
    ).all()

    vector_norms = np.linalg.norm(
        cached_vectors,
        axis=1,
    )

    assert (
        vector_norms > 0
    ).all()

    return cached_vectors


# =========================================================
# Discover existing vs pending Embedding batches
# =========================================================

existing_embedding_cache_count = 0
pending_embedding_batches = []


for batch_records in embedding_batches:

    cache_path = (
        embedding_cache_path_for_batch(
            batch_records
        )
    )

    if cache_path.exists():

        load_and_validate_embedding_cache(
            batch_records,
            cache_path,
        )

        existing_embedding_cache_count += 1

    else:

        pending_embedding_batches.append(
            batch_records
        )


# =========================================================
# Header / progress
# =========================================================

print(
    "=== NB12i SEMANTIC DEDUP EMBEDDING ==="
)

print()

print(
    "Embedding Model:",
    resolved_embedding_model_id,
)

print(
    "Exact-unique Claim texts:",
    len(embedding_text_records),
)

print(
    "Embedding Batch size:",
    EMBEDDING_BATCH_SIZE,
)

print(
    "Total Embedding batches:",
    len(embedding_batches),
)

print(
    "Validated caches before run:",
    existing_embedding_cache_count,
)

print(
    "Pending Embedding batches:",
    len(pending_embedding_batches),
)

print(
    "Internal request rate:",
    EMBEDDING_INTERNAL_RPM,
    "requests/minute",
)


if pending_embedding_batches:

    print(
        "Approximate minimum request-start time:",
        round(
            len(
                pending_embedding_batches
            )
            / float(
                EMBEDDING_INTERNAL_RPM
            ),
            1,
        ),
        "minutes",
    )


print()


# =========================================================
# Current-run telemetry
# =========================================================

embedding_api_calls_this_run = 0
embedding_retry_events_this_run = 0
embedding_new_caches_this_run = 0


# =========================================================
# Resolve endpoint/key ONLY if API work is really pending
# =========================================================

arvan_embedding_client_runtime = None


def extract_env_name_from_expression(
    expression,
):

    markers = [
        'os.getenv("',
        "os.getenv('",
        'os.environ.get("',
        "os.environ.get('",
    ]

    for marker in markers:

        marker_index = (
            expression.find(marker)
        )

        if marker_index < 0:
            continue

        start = (
            marker_index
            + len(marker)
        )

        quote = (
            '"'
            if '"' in marker
            else "'"
        )

        end = expression.find(
            quote,
            start,
        )

        if end > start:
            return expression[
                start:end
            ]

    return None


if pending_embedding_batches:

    arvan_embedding_url_runtime = (
        os.getenv(
            "ARVAN_EMBEDDING_URL"
        )
    )

    assert (
        arvan_embedding_url_runtime
    ), (
        "ARVAN_EMBEDDING_URL is not available "
        "in the current environment."
    )


    embedding_api_key_expression = None

    for source in reference_sources:

        if (
            "arvan_embedding_client = OpenAI("
            not in source
        ):
            continue

        lines = source.splitlines()

        client_start = None

        for line_index, raw_line in enumerate(
            lines
        ):

            if (
                "arvan_embedding_client = OpenAI("
                in raw_line
            ):
                client_start = (
                    line_index
                )
                break

        if client_start is None:
            continue

        for raw_line in lines[
            client_start:
            client_start + 12
        ]:

            stripped = raw_line.strip()

            if stripped.startswith(
                "api_key="
            ):

                embedding_api_key_expression = (
                    stripped
                    .split("=", 1)[1]
                    .rstrip(",")
                    .strip()
                )

                break

        if embedding_api_key_expression:
            break


    assert (
        embedding_api_key_expression
    ), (
        "Could not locate Notebook-13 "
        "Embedding api_key expression."
    )


    embedding_api_key_env_name = (
        extract_env_name_from_expression(
            embedding_api_key_expression
        )
    )


    if (
        embedding_api_key_env_name
        is None
    ):

        possible_variable_name = (
            embedding_api_key_expression
        )

        if (
            possible_variable_name
            .isidentifier()
        ):

            prefix = (
                possible_variable_name
                + " ="
            )

            for raw_line in reference_lines:

                stripped = (
                    raw_line.strip()
                )

                if not stripped.startswith(
                    prefix
                ):
                    continue

                rhs = (
                    stripped
                    .split("=", 1)[1]
                    .strip()
                )

                discovered = (
                    extract_env_name_from_expression(
                        rhs
                    )
                )

                if discovered:

                    embedding_api_key_env_name = (
                        discovered
                    )

                    break


    assert (
        embedding_api_key_env_name
    ), (
        "Could not safely resolve the API-key "
        "environment variable used by Notebook 13."
    )


    embedding_api_key_value = (
        os.getenv(
            embedding_api_key_env_name
        )
    )


    assert (
        embedding_api_key_value
    ), (
        "Notebook-13 Embedding API-key "
        "environment variable is not set: "
        f"{embedding_api_key_env_name}"
    )


    arvan_embedding_client_runtime = (
        OpenAI(
            api_key=(
                embedding_api_key_value
            ),
            base_url=(
                arvan_embedding_url_runtime
            ),
            timeout=180.0,
        )
    )


# =========================================================
# Request pacing
# =========================================================

embedding_request_interval_seconds = (
    60.0
    / float(
        EMBEDDING_INTERNAL_RPM
    )
)

embedding_next_request_time = (
    time.monotonic()
)


def wait_for_embedding_slot():

    global embedding_next_request_time

    now = time.monotonic()

    wait_seconds = max(
        0.0,
        embedding_next_request_time
        - now,
    )

    if wait_seconds > 0:
        time.sleep(
            wait_seconds
        )

    actual_start = (
        time.monotonic()
    )

    embedding_next_request_time = (
        max(
            actual_start,
            embedding_next_request_time,
        )
        + embedding_request_interval_seconds
    )


# =========================================================
# Embed ONLY missing batches
# =========================================================

for batch_number, batch_records in enumerate(
    pending_embedding_batches,
    start=1,
):

    batch_id = embedding_batch_id(
        batch_records
    )

    cache_path = (
        embedding_cache_path_for_batch(
            batch_records
        )
    )

    texts = [
        item["claim_text_fa"]
        for item in batch_records
    ]

    embedding_vectors = None
    last_error = None


    for attempt in range(
        1,
        EMBEDDING_MAX_ATTEMPTS + 1,
    ):

        try:

            wait_for_embedding_slot()

            embedding_api_calls_this_run += 1


            response = (
                arvan_embedding_client_runtime
                .embeddings
                .create(
                    model=(
                        resolved_embedding_model_id
                    ),
                    input=texts,
                )
            )


            response_items = sorted(
                response.data,
                key=lambda item: (
                    item.index
                ),
            )


            assert (
                len(response_items)
                == len(texts)
            ), (
                "Embedding response count "
                "does not match request count."
            )


            embedding_vectors = np.asarray(
                [
                    item.embedding
                    for item in response_items
                ],
                dtype=np.float32,
            )


            assert (
                embedding_vectors.ndim == 2
            )

            assert (
                embedding_vectors.shape[0]
                == len(texts)
            )

            assert np.isfinite(
                embedding_vectors
            ).all()


            norms = np.linalg.norm(
                embedding_vectors,
                axis=1,
            )


            assert (
                norms > 0
            ).all()


            embedding_vectors = (
                embedding_vectors
                / norms[:, None]
            ).astype(
                np.float32
            )


            break


        except Exception as exc:

            last_error = (
                f"{type(exc).__name__}: "
                f"{exc}"
            )

            embedding_retry_events_this_run += 1


            if (
                attempt
                >= EMBEDDING_MAX_ATTEMPTS
            ):

                raise RuntimeError(
                    "Embedding batch failed after "
                    f"{EMBEDDING_MAX_ATTEMPTS} attempts: "
                    f"{batch_id} | "
                    f"{last_error}"
                ) from exc


            time.sleep(
                min(
                    60.0,
                    10.0
                    * float(attempt),
                )
            )


    assert (
        embedding_vectors
        is not None
    )


    # No overwrite.
    with cache_path.open(
        "xb"
    ) as file:

        np.savez_compressed(
            file,
            cache_version=np.asarray(
                EMBEDDING_CACHE_VERSION
            ),
            model_id=np.asarray(
                resolved_embedding_model_id
            ),
            batch_id=np.asarray(
                batch_id
            ),
            text_ids=np.asarray(
                [
                    item["text_id"]
                    for item in batch_records
                ]
            ),
            vectors=(
                embedding_vectors
            ),
        )


    load_and_validate_embedding_cache(
        batch_records,
        cache_path,
    )


    embedding_new_caches_this_run += 1


    if (
        batch_number % 10 == 0
        or batch_number
        == len(
            pending_embedding_batches
        )
    ):

        print(
            "[",
            batch_number,
            "/",
            len(
                pending_embedding_batches
            ),
            "] VALID |",
            batch_id,
        )


# =========================================================
# Reload and validate ALL Embedding caches
# =========================================================

all_embedding_blocks = []


for batch_records in embedding_batches:

    cache_path = (
        embedding_cache_path_for_batch(
            batch_records
        )
    )

    assert (
        cache_path.exists()
    ), (
        "Embedding cache missing after run: "
        f"{cache_path.name}"
    )


    block_vectors = (
        load_and_validate_embedding_cache(
            batch_records,
            cache_path,
        )
    )

    all_embedding_blocks.append(
        block_vectors
    )


embedding_matrix = np.vstack(
    all_embedding_blocks
).astype(
    np.float32
)


assert (
    embedding_matrix.shape[0]
    == len(
        embedding_text_records
    )
)


embedding_dimension = (
    embedding_matrix.shape[1]
)


assert (
    embedding_dimension > 0
)

assert np.isfinite(
    embedding_matrix
).all()


matrix_norms = np.linalg.norm(
    embedding_matrix,
    axis=1,
)


assert np.allclose(
    matrix_norms,
    1.0,
    atol=1e-5,
)


# =========================================================
# Freeze / reuse consolidated Embedding matrix
# =========================================================

if embedding_matrix_path.exists():

    with np.load(
        embedding_matrix_path,
        allow_pickle=False,
    ) as payload:

        disk_model = str(
            payload[
                "model_id"
            ].item()
        )

        disk_text_ids = [
            str(value)
            for value in payload[
                "text_ids"
            ].tolist()
        ]

        disk_matrix = np.asarray(
            payload[
                "vectors"
            ],
            dtype=np.float32,
        )


    assert (
        disk_model
        == resolved_embedding_model_id
    )

    assert (
        disk_text_ids
        == embedding_text_ids
    )

    assert np.array_equal(
        disk_matrix,
        embedding_matrix,
    )


    embedding_matrix_write_status = (
        "REUSED_EXISTING_IDENTICAL"
    )


else:

    with embedding_matrix_path.open(
        "xb"
    ) as file:

        np.savez_compressed(
            file,
            version=np.asarray(
                EMBEDDING_STAGE_VERSION
            ),
            model_id=np.asarray(
                resolved_embedding_model_id
            ),
            text_ids=np.asarray(
                embedding_text_ids
            ),
            vectors=(
                embedding_matrix
            ),
        )


    embedding_matrix_write_status = (
        "CREATED_NEW"
    )


embedding_matrix_sha256 = (
    file_sha256(
        embedding_matrix_path
    )
)


# =========================================================
# Candidate artifact:
#
# If BOTH candidate artifact and frozen report already
# exist, verify the SHA chain and reuse the artifact.
#
# This avoids recomputing all cosine neighborhoods on
# ordinary reruns.
# =========================================================

reuse_existing_candidate_artifact = (
    semantic_neighbor_candidates_path.exists()
    and existing_semantic_candidate_report
    is not None
)


if reuse_existing_candidate_artifact:

    semantic_neighbor_candidates = (
        load_jsonl(
            semantic_neighbor_candidates_path
        )
    )


    semantic_neighbor_candidates_sha256 = (
        file_sha256(
            semantic_neighbor_candidates_path
        )
    )


    assert (
        semantic_neighbor_candidates_sha256
        == existing_semantic_candidate_report[
            "artifacts"
        ][
            "neighbor_candidates_sha256"
        ]
    ), (
        "Existing Candidate artifact SHA-256 "
        "does not match frozen report."
    )


    assert (
        existing_semantic_candidate_report[
            "artifacts"
        ][
            "embedding_matrix_sha256"
        ]
        == embedding_matrix_sha256
    ), (
        "Existing frozen report references "
        "a different Embedding matrix."
    )


    assert (
        existing_semantic_candidate_report[
            "embedding_model"
        ]
        == resolved_embedding_model_id
    )


    assert (
        existing_semantic_candidate_report[
            "top_k"
        ]
        == min(
            SEMANTIC_NEIGHBOR_TOP_K,
            len(
                embedding_text_records
            )
            - 1,
        )
    )


else:

    # -----------------------------------------------------
    # Local blockwise Top-K cosine generation.
    # No API calls.
    # -----------------------------------------------------

    neighbor_pair_directions = {}

    text_count = len(
        embedding_text_records
    )

    effective_top_k = min(
        SEMANTIC_NEIGHBOR_TOP_K,
        text_count - 1,
    )


    for block_start in range(
        0,
        text_count,
        SIMILARITY_BLOCK_SIZE,
    ):

        block_end = min(
            block_start
            + SIMILARITY_BLOCK_SIZE,
            text_count,
        )


        block_matrix = (
            embedding_matrix[
                block_start:
                block_end
            ]
        )


        similarities = (
            block_matrix
            @ embedding_matrix.T
        )


        for local_row_index in range(
            block_end
            - block_start
        ):

            global_row_index = (
                block_start
                + local_row_index
            )

            row = similarities[
                local_row_index
            ]

            row[
                global_row_index
            ] = -np.inf


            neighbor_indices = (
                np.argpartition(
                    row,
                    -effective_top_k,
                )[
                    -effective_top_k:
                ]
            )


            neighbor_indices = sorted(
                neighbor_indices.tolist(),
                key=lambda index: (
                    -float(
                        row[index]
                    ),
                    embedding_text_ids[
                        index
                    ],
                ),
            )


            for neighbor_index in (
                neighbor_indices
            ):

                similarity = float(
                    row[
                        neighbor_index
                    ]
                )


                left_index = min(
                    global_row_index,
                    neighbor_index,
                )

                right_index = max(
                    global_row_index,
                    neighbor_index,
                )


                pair_key = (
                    left_index,
                    right_index,
                )


                if (
                    pair_key
                    not in neighbor_pair_directions
                ):

                    neighbor_pair_directions[
                        pair_key
                    ] = {
                        "similarity": similarity,
                        "directions": set(),
                    }


                else:

                    assert np.isclose(
                        neighbor_pair_directions[
                            pair_key
                        ][
                            "similarity"
                        ],
                        similarity,
                        atol=1e-6,
                    )


                neighbor_pair_directions[
                    pair_key
                ][
                    "directions"
                ].add(
                    global_row_index
                )


    # -----------------------------------------------------
    # Materialize unique pairs
    # -----------------------------------------------------

    semantic_neighbor_candidates = []


    for (
        left_index,
        right_index,
    ), pair_info in (
        neighbor_pair_directions.items()
    ):

        left_record = (
            embedding_text_records[
                left_index
            ]
        )

        right_record = (
            embedding_text_records[
                right_index
            ]
        )


        assert (
            left_record[
                "claim_text_fa"
            ]
            != right_record[
                "claim_text_fa"
            ]
        ), (
            "Exact duplicate text leaked into "
            "Semantic Candidate Generation."
        )


        pair_material = (
            left_record["text_id"]
            + "\n"
            + right_record["text_id"]
        )


        pair_sha256 = hashlib.sha256(
            pair_material.encode(
                "utf-8"
            )
        ).hexdigest()


        semantic_neighbor_candidates.append({
            "candidate_pair_id": (
                "semantic-candidate:"
                + pair_sha256[:24]
            ),

            "text_id_a": (
                left_record[
                    "text_id"
                ]
            ),

            "text_id_b": (
                right_record[
                    "text_id"
                ]
            ),

            "raw_claim_ids_a": (
                left_record[
                    "raw_claim_ids"
                ]
            ),

            "raw_claim_ids_b": (
                right_record[
                    "raw_claim_ids"
                ]
            ),

            "cosine_similarity": round(
                pair_info[
                    "similarity"
                ],
                8,
            ),

            "mutual_top_k": (
                len(
                    pair_info[
                        "directions"
                    ]
                )
                == 2
            ),

            "top_k": (
                effective_top_k
            ),

            "embedding_model": (
                resolved_embedding_model_id
            ),

            "candidate_generation_version": (
                EMBEDDING_STAGE_VERSION
            ),
        })


    semantic_neighbor_candidates = sorted(
        semantic_neighbor_candidates,
        key=lambda item: (
            -item[
                "cosine_similarity"
            ],
            item[
                "candidate_pair_id"
            ],
        ),
    )


# =========================================================
# Validate loaded OR generated Candidate artifact
# =========================================================

effective_top_k = min(
    SEMANTIC_NEIGHBOR_TOP_K,
    len(
        embedding_text_records
    )
    - 1,
)


candidate_pair_ids = [
    item["candidate_pair_id"]
    for item in (
        semantic_neighbor_candidates
    )
]


assert (
    len(candidate_pair_ids)
    == len(set(candidate_pair_ids))
)


valid_text_ids = set(
    embedding_text_ids
)


for candidate in (
    semantic_neighbor_candidates
):

    assert (
        candidate["text_id_a"]
        in valid_text_ids
    )

    assert (
        candidate["text_id_b"]
        in valid_text_ids
    )

    assert (
        candidate["text_id_a"]
        != candidate["text_id_b"]
    )

    assert (
        candidate["top_k"]
        == effective_top_k
    )

    assert (
        candidate["embedding_model"]
        == resolved_embedding_model_id
    )

    assert (
        candidate[
            "candidate_generation_version"
        ]
        == EMBEDDING_STAGE_VERSION
    )


# =========================================================
# Canonical Candidate serialization
# =========================================================

semantic_neighbor_candidates_text = "".join(
    json.dumps(
        item,
        ensure_ascii=False,
        sort_keys=True,
    )
    + "\n"
    for item in (
        semantic_neighbor_candidates
    )
)


semantic_neighbor_candidates_sha256 = (
    hashlib.sha256(
        semantic_neighbor_candidates_text.encode(
            "utf-8"
        )
    ).hexdigest()
)


semantic_neighbor_candidates_write_status = (
    publish_frozen_artifact(
        semantic_neighbor_candidates_path,
        semantic_neighbor_candidates_text,
    )
)


assert (
    file_sha256(
        semantic_neighbor_candidates_path
    )
    == semantic_neighbor_candidates_sha256
)


# =========================================================
# Diagnostics from actual Candidate artifact
# =========================================================

candidate_similarity_values = np.asarray(
    [
        item[
            "cosine_similarity"
        ]
        for item in (
            semantic_neighbor_candidates
        )
    ],
    dtype=np.float64,
)


assert (
    candidate_similarity_values.size > 0
)


similarity_percentiles = {
    "p50": float(
        np.percentile(
            candidate_similarity_values,
            50,
        )
    ),

    "p75": float(
        np.percentile(
            candidate_similarity_values,
            75,
        )
    ),

    "p90": float(
        np.percentile(
            candidate_similarity_values,
            90,
        )
    ),

    "p95": float(
        np.percentile(
            candidate_similarity_values,
            95,
        )
    ),

    "p97": float(
        np.percentile(
            candidate_similarity_values,
            97,
        )
    ),

    "p99": float(
        np.percentile(
            candidate_similarity_values,
            99,
        )
    ),

    "max": float(
        candidate_similarity_values.max()
    ),
}


diagnostic_similarity_counts = {
    threshold: int(
        np.sum(
            candidate_similarity_values
            >= threshold
        )
    )
    for threshold in (
        0.70,
        0.75,
        0.80,
        0.85,
        0.90,
        0.95,
    )
}


mutual_candidate_count = sum(
    1
    for item in (
        semantic_neighbor_candidates
    )
    if item[
        "mutual_top_k"
    ]
)


# =========================================================
# Build the report for THIS runtime state
# =========================================================

semantic_candidate_report_current = {
    "report_version": (
        "claim_semantic_candidate_generation_report_v1"
    ),

    "candidate_generation_version": (
        EMBEDDING_STAGE_VERSION
    ),

    "source_validated_claims": (
        validated_claims_path.name
    ),

    "source_validated_claims_sha256": (
        validated_claims_sha256
    ),

    "source_exact_dedup_report": (
        dedup_stage1_report_path.name
    ),

    "embedding_model": (
        resolved_embedding_model_id
    ),

    "embedding_texts": (
        len(
            embedding_text_records
        )
    ),

    "embedding_dimension": (
        embedding_dimension
    ),

    "embedding_batch_size": (
        EMBEDDING_BATCH_SIZE
    ),

    "embedding_batches": (
        len(
            embedding_batches
        )
    ),

    # Run-local telemetry.
    "embedding_api_calls_this_run": (
        embedding_api_calls_this_run
    ),

    "embedding_retry_events_this_run": (
        embedding_retry_events_this_run
    ),

    "embedding_new_caches_this_run": (
        embedding_new_caches_this_run
    ),

    "top_k": (
        effective_top_k
    ),

    "unique_neighbor_candidate_pairs": (
        len(
            semantic_neighbor_candidates
        )
    ),

    "mutual_top_k_pairs": (
        mutual_candidate_count
    ),

    "similarity_percentiles": {
        key: round(
            value,
            8,
        )
        for key, value in (
            similarity_percentiles.items()
        )
    },

    "diagnostic_similarity_counts": {
        f">={threshold:.2f}": count
        for threshold, count in (
            diagnostic_similarity_counts.items()
        )
    },

    "semantic_relation_decisions_made": 0,

    "llm_calls": 0,

    "automatic_merge_applied": False,

    "claim_ids_assigned": False,

    "next_stage": (
        "SEMANTIC_CANDIDATE_SELECTION"
    ),

    "artifacts": {
        "embedding_text_index": (
            embedding_text_index_path.name
        ),

        "embedding_text_index_sha256": (
            embedding_text_index_sha256
        ),

        "embedding_matrix": (
            embedding_matrix_path.name
        ),

        "embedding_matrix_sha256": (
            embedding_matrix_sha256
        ),

        "neighbor_candidates": (
            semantic_neighbor_candidates_path.name
        ),

        "neighbor_candidates_sha256": (
            semantic_neighbor_candidates_sha256
        ),
    },
}


# =========================================================
# CRITICAL RERUN-SAFETY FIX
#
# These three fields legitimately change between:
#
# First run:
#   API calls = 63
#   new caches = 63
#
# Cache-only rerun:
#   API calls = 0
#   new caches = 0
#
# They must NOT invalidate the already-frozen report.
#
# Therefore:
# 1. compare all deterministic fields;
# 2. require them to be identical;
# 3. preserve the original creation-run telemetry;
# 4. republish only byte-identical content.
# =========================================================

RUN_LOCAL_REPORT_FIELDS = {
    "embedding_api_calls_this_run",
    "embedding_retry_events_this_run",
    "embedding_new_caches_this_run",
}


report_creation_telemetry_preserved = False


if (
    existing_semantic_candidate_report
    is not None
):

    current_stable_report = {
        key: value
        for key, value in (
            semantic_candidate_report_current.items()
        )
        if key not in (
            RUN_LOCAL_REPORT_FIELDS
        )
    }


    existing_stable_report = {
        key: value
        for key, value in (
            existing_semantic_candidate_report.items()
        )
        if key not in (
            RUN_LOCAL_REPORT_FIELDS
        )
    }


    assert (
        current_stable_report
        == existing_stable_report
    ), (
        "Existing frozen Semantic Candidate "
        "report differs in deterministic content. "
        "Refusing reuse."
    )


    for field in (
        RUN_LOCAL_REPORT_FIELDS
    ):

        assert (
            field
            in existing_semantic_candidate_report
        ), (
            "Existing frozen report is missing "
            f"run-local field: {field}"
        )


    semantic_candidate_report = dict(
        existing_semantic_candidate_report
    )


    report_creation_telemetry_preserved = (
        True
    )


else:

    semantic_candidate_report = (
        semantic_candidate_report_current
    )


# =========================================================
# Canonical frozen Report serialization
# =========================================================

semantic_candidate_report_text = (
    json.dumps(
        semantic_candidate_report,
        ensure_ascii=False,
        indent=2,
        sort_keys=True,
    )
    + "\n"
)


semantic_candidate_report_write_status = (
    publish_frozen_artifact(
        semantic_candidate_report_path,
        semantic_candidate_report_text,
    )
)


# =========================================================
# Independent disk revalidation
# =========================================================

assert (
    load_jsonl(
        embedding_text_index_path
    )
    == embedding_text_records
)


assert (
    load_jsonl(
        semantic_neighbor_candidates_path
    )
    == semantic_neighbor_candidates
)


assert (
    load_json(
        semantic_candidate_report_path
    )
    == semantic_candidate_report
)


assert (
    file_sha256(
        semantic_neighbor_candidates_path
    )
    == semantic_neighbor_candidates_sha256
)


assert (
    file_sha256(
        embedding_matrix_path
    )
    == embedding_matrix_sha256
)


assert (
    semantic_candidate_report[
        "artifacts"
    ][
        "neighbor_candidates_sha256"
    ]
    == semantic_neighbor_candidates_sha256
)


assert (
    semantic_candidate_report[
        "artifacts"
    ][
        "embedding_matrix_sha256"
    ]
    == embedding_matrix_sha256
)


# =========================================================
# Gate
# =========================================================

semantic_candidate_generation_status = (
    "PASS"
)

semantic_candidate_selection_readiness = (
    "READY_FOR_CANDIDATE_SELECTION"
)


# =========================================================
# Result
# =========================================================

print()

print(
    "=== NB12i SEMANTIC DEDUP "
    "CANDIDATE GENERATION ==="
)

print()

print(
    "Exact-unique Claim texts embedded:",
    len(
        embedding_text_records
    ),
)

print(
    "Embedding dimension:",
    embedding_dimension,
)

print(
    "Embedding batches:",
    len(
        embedding_batches
    ),
)

print(
    "Embedding API calls THIS run:",
    embedding_api_calls_this_run,
)

print(
    "Embedding retries THIS run:",
    embedding_retry_events_this_run,
)

print(
    "New Embedding caches THIS run:",
    embedding_new_caches_this_run,
)


print()

print(
    "Top-K neighbors per text:",
    effective_top_k,
)

print(
    "Unique semantic neighbor pairs:",
    len(
        semantic_neighbor_candidates
    ),
)

print(
    "Mutual Top-K pairs:",
    mutual_candidate_count,
)


print()

print(
    "Similarity percentiles:"
)

for key in [
    "p50",
    "p75",
    "p90",
    "p95",
    "p97",
    "p99",
    "max",
]:

    print(
        " ",
        key,
        "=",
        round(
            similarity_percentiles[
                key
            ],
            6,
        ),
    )


print()

print(
    "Diagnostic pair counts:"
)

for threshold in [
    0.70,
    0.75,
    0.80,
    0.85,
    0.90,
    0.95,
]:

    print(
        "  similarity >=",
        threshold,
        ":",
        diagnostic_similarity_counts[
            threshold
        ],
    )


print()

print(
    "Embedding Matrix SHA-256:",
    embedding_matrix_sha256,
)

print(
    "Matrix write status:",
    embedding_matrix_write_status,
)


print()

print(
    "Candidate artifact SHA-256:",
    semantic_neighbor_candidates_sha256,
)

print(
    "Candidate write status:",
    semantic_neighbor_candidates_write_status,
)


print()

print(
    "Frozen report creation telemetry preserved:",
    report_creation_telemetry_preserved,
)

print(
    "Frozen report API calls:",
    semantic_candidate_report[
        "embedding_api_calls_this_run"
    ],
)

print(
    "Frozen report new caches:",
    semantic_candidate_report[
        "embedding_new_caches_this_run"
    ],
)

print(
    "Report write status:",
    semantic_candidate_report_write_status,
)


print()

print(
    "LLM calls performed:",
    0,
)

print(
    "Semantic relation decisions made:",
    0,
)

print(
    "Automatic merge applied:",
    False,
)

print(
    "Claim IDs assigned:",
    False,
)


print()

print(
    "Next stage:",
    "SEMANTIC_CANDIDATE_SELECTION",
)

print(
    "Candidate readiness:",
    semantic_candidate_selection_readiness,
)


print()

print(
    "CODE CELL 22 GATE:",
    semantic_candidate_generation_status,
)
```

<div style="direction: rtl; white-space: normal; line-height: 1.7;">

<h2>انتخاب Candidateهای Semantic Dedup برای داوری</h2>

<p>
تولید Candidate با Embedding با موفقیت پایان یافته است.
از میان ۱۵۱٬۴۹۳ Pair همسایه، اکنون فقط Pairهایی انتخاب می‌شوند
که شباهت آن‌ها برای ورود به داوری Semantic Dedup به‌اندازه کافی بالا باشد.
</p>

<p>
سیاست انتخاب بر اساس توزیع واقعی Similarity همین Registry تعیین می‌شود:
تمام Pairهای با Similarity حداقل ۰٫۷۵ انتخاب می‌شوند؛
همچنین Pairهای بازه ۰٫۷۰ تا ۰٫۷۵ فقط زمانی انتخاب می‌شوند
که رابطه همسایگی آن‌ها Mutual Top-K باشد.
</p>

<ul>
<li>هیچ LLM یا API Call انجام نمی‌شود.</li>
<li>هیچ Claim حذف یا Merge نمی‌شود.</li>
<li>هیچ Relation معنایی هنوز قطعی نمی‌شود.</li>
<li>Exact Duplicateهای مرحله قبل وارد این مرحله نمی‌شوند.</li>
<li>خروجی این مرحله فقط Queue نهایی برای Semantic Judge است.</li>
</ul>

</div>

```python
# Code Cell 23

assert (
    semantic_candidate_generation_status
    == "PASS"
), (
    "Code Cell 22 must PASS before "
    "Semantic Candidate Selection."
)

assert (
    semantic_candidate_selection_readiness
    == "READY_FOR_CANDIDATE_SELECTION"
)


SEMANTIC_SELECTION_VERSION = (
    "claim_semantic_candidate_selection_v1"
)

SEMANTIC_PRIMARY_THRESHOLD = 0.75

SEMANTIC_MUTUAL_RECOVERY_THRESHOLD = 0.70


semantic_selected_candidates_path = (
    registry_artifact_root
    / "claim_semantic_dedup_candidates_selected_v1.jsonl"
)

semantic_selection_report_path = (
    registry_artifact_root
    / "claim_semantic_candidate_selection_report_v1.json"
)


# ---------------------------------------------------------
# Independently reload candidate-generation artifacts.
# ---------------------------------------------------------

selection_candidates = load_jsonl(
    semantic_neighbor_candidates_path
)

selection_text_index = load_jsonl(
    embedding_text_index_path
)

selection_candidate_report = load_json(
    semantic_candidate_report_path
)


assert (
    file_sha256(
        semantic_neighbor_candidates_path
    )
    == semantic_neighbor_candidates_sha256
), (
    "Semantic neighbor candidate artifact changed."
)


assert (
    len(
        selection_candidates
    )
    == selection_candidate_report[
        "unique_neighbor_candidate_pairs"
    ]
)


assert (
    selection_candidate_report[
        "semantic_relation_decisions_made"
    ]
    == 0
)


# ---------------------------------------------------------
# Build exact text lookup.
# ---------------------------------------------------------

selection_text_by_id = {
    item[
        "text_id"
    ]: item
    for item in (
        selection_text_index
    )
}


assert (
    len(
        selection_text_by_id
    )
    == len(
        selection_text_index
    )
)


# ---------------------------------------------------------
# Rebuild Raw Claim metadata lookup.
# ---------------------------------------------------------

selection_claims = load_jsonl(
    validated_claims_path
)

selection_claim_by_id = {
    item[
        "raw_claim_id"
    ]: item
    for item in (
        selection_claims
    )
}


assert (
    len(
        selection_claim_by_id
    )
    == len(
        selection_claims
    )
)


selection_evidence = load_jsonl(
    full_evidence_path
)

selection_source_types_by_claim = defaultdict(
    set
)


for evidence in (
    selection_evidence
):

    raw_claim_id = evidence[
        "raw_claim_id"
    ]


    if (
        raw_claim_id
        in selection_claim_by_id
    ):

        selection_source_types_by_claim[
            raw_claim_id
        ].add(
            evidence[
                "source_type"
            ]
        )


# ---------------------------------------------------------
# Deterministic selection policy.
#
# Channel A:
#   similarity >= 0.75
#
# Channel B:
#   0.70 <= similarity < 0.75
#   AND mutual_top_k == True
#
# No LLM.
# No relation decision.
# ---------------------------------------------------------

semantic_selected_candidates = []


for candidate in (
    selection_candidates
):

    similarity = float(
        candidate[
            "cosine_similarity"
        ]
    )

    mutual_top_k = bool(
        candidate[
            "mutual_top_k"
        ]
    )


    selected_by = []


    if (
        similarity
        >= SEMANTIC_PRIMARY_THRESHOLD
    ):

        selected_by.append(
            "HIGH_SIMILARITY"
        )


    elif (
        similarity
        >= SEMANTIC_MUTUAL_RECOVERY_THRESHOLD
        and mutual_top_k
    ):

        selected_by.append(
            "MUTUAL_TOP_K_RECOVERY"
        )


    if not selected_by:
        continue


    text_a = (
        selection_text_by_id[
            candidate[
                "text_id_a"
            ]
        ]
    )

    text_b = (
        selection_text_by_id[
            candidate[
                "text_id_b"
            ]
        ]
    )


    raw_claim_ids_a = (
        candidate[
            "raw_claim_ids_a"
        ]
    )

    raw_claim_ids_b = (
        candidate[
            "raw_claim_ids_b"
        ]
    )


    source_types_a = sorted({
        source_type
        for raw_claim_id in (
            raw_claim_ids_a
        )
        for source_type in (
            selection_source_types_by_claim[
                raw_claim_id
            ]
        )
    })


    source_types_b = sorted({
        source_type
        for raw_claim_id in (
            raw_claim_ids_b
        )
        for source_type in (
            selection_source_types_by_claim[
                raw_claim_id
            ]
        )
    })


    claim_types_a = sorted({
        selection_claim_by_id[
            raw_claim_id
        ][
            "claim_type"
        ]
        for raw_claim_id in (
            raw_claim_ids_a
        )
    })


    claim_types_b = sorted({
        selection_claim_by_id[
            raw_claim_id
        ][
            "claim_type"
        ]
        for raw_claim_id in (
            raw_claim_ids_b
        )
    })


    authority_classes_a = sorted({
        selection_claim_by_id[
            raw_claim_id
        ][
            "authority_class"
        ]
        for raw_claim_id in (
            raw_claim_ids_a
        )
    })


    authority_classes_b = sorted({
        selection_claim_by_id[
            raw_claim_id
        ][
            "authority_class"
        ]
        for raw_claim_id in (
            raw_claim_ids_b
        )
    })


    if (
        similarity
        >= 0.90
    ):

        similarity_band = "GE_0_90"

    elif (
        similarity
        >= 0.85
    ):

        similarity_band = "0_85_TO_0_90"

    elif (
        similarity
        >= 0.80
    ):

        similarity_band = "0_80_TO_0_85"

    elif (
        similarity
        >= 0.75
    ):

        similarity_band = "0_75_TO_0_80"

    else:

        similarity_band = "0_70_TO_0_75_MUTUAL"


    semantic_selected_candidates.append({
        "candidate_pair_id": (
            candidate[
                "candidate_pair_id"
            ]
        ),

        "text_id_a": (
            candidate[
                "text_id_a"
            ]
        ),

        "text_id_b": (
            candidate[
                "text_id_b"
            ]
        ),

        "claim_text_a": (
            text_a[
                "claim_text_fa"
            ]
        ),

        "claim_text_b": (
            text_b[
                "claim_text_fa"
            ]
        ),

        "raw_claim_ids_a": (
            raw_claim_ids_a
        ),

        "raw_claim_ids_b": (
            raw_claim_ids_b
        ),

        "claim_types_a": (
            claim_types_a
        ),

        "claim_types_b": (
            claim_types_b
        ),

        "source_types_a": (
            source_types_a
        ),

        "source_types_b": (
            source_types_b
        ),

        "authority_classes_a": (
            authority_classes_a
        ),

        "authority_classes_b": (
            authority_classes_b
        ),

        "cosine_similarity": (
            similarity
        ),

        "mutual_top_k": (
            mutual_top_k
        ),

        "similarity_band": (
            similarity_band
        ),

        "selected_by": (
            selected_by
        ),

        "selection_version": (
            SEMANTIC_SELECTION_VERSION
        ),

        "semantic_relation_decision": None,
    })


semantic_selected_candidates = sorted(
    semantic_selected_candidates,
    key=lambda item: (
        -item[
            "cosine_similarity"
        ],
        item[
            "candidate_pair_id"
        ],
    ),
)


# ---------------------------------------------------------
# Structural validation.
# ---------------------------------------------------------

selected_pair_ids = [
    item[
        "candidate_pair_id"
    ]
    for item in (
        semantic_selected_candidates
    )
]


assert (
    len(
        selected_pair_ids
    )
    == len(
        set(
            selected_pair_ids
        )
    )
)


assert all(
    item[
        "claim_text_a"
    ]
    != item[
        "claim_text_b"
    ]
    for item in (
        semantic_selected_candidates
    )
), (
    "Exact duplicate text leaked into "
    "Semantic Judge candidate queue."
)


assert all(
    (
        item[
            "cosine_similarity"
        ]
        >= SEMANTIC_PRIMARY_THRESHOLD
    )
    or (
        item[
            "cosine_similarity"
        ]
        >= SEMANTIC_MUTUAL_RECOVERY_THRESHOLD
        and item[
            "mutual_top_k"
        ]
    )
    for item in (
        semantic_selected_candidates
    )
)


# ---------------------------------------------------------
# Selection diagnostics.
# ---------------------------------------------------------

selection_channel_counts = Counter()

selection_band_counts = Counter()

selection_source_pair_counts = Counter()

selection_claim_type_pair_counts = Counter()


for item in (
    semantic_selected_candidates
):

    for channel in (
        item[
            "selected_by"
        ]
    ):

        selection_channel_counts[
            channel
        ] += 1


    selection_band_counts[
        item[
            "similarity_band"
        ]
    ] += 1


    source_pair_key = (
        " + ".join(
            item[
                "source_types_a"
            ]
        )
        + "  <->  "
        + " + ".join(
            item[
                "source_types_b"
            ]
        )
    )


    selection_source_pair_counts[
        source_pair_key
    ] += 1


    claim_type_pair_key = (
        " + ".join(
            item[
                "claim_types_a"
            ]
        )
        + "  <->  "
        + " + ".join(
            item[
                "claim_types_b"
            ]
        )
    )


    selection_claim_type_pair_counts[
        claim_type_pair_key
    ] += 1


cross_source_selected_count = sum(
    1
    for item in (
        semantic_selected_candidates
    )
    if (
        set(
            item[
                "source_types_a"
            ]
        )
        != set(
            item[
                "source_types_b"
            ]
        )
    )
)


mutual_selected_count = sum(
    1
    for item in (
        semantic_selected_candidates
    )
    if item[
        "mutual_top_k"
    ]
)


selected_similarity_values = np.asarray(
    [
        item[
            "cosine_similarity"
        ]
        for item in (
            semantic_selected_candidates
        )
    ],
    dtype=np.float64,
)


assert (
    selected_similarity_values.size
    > 0
)


selected_similarity_stats = {
    "min": float(
        selected_similarity_values.min()
    ),

    "p25": float(
        np.percentile(
            selected_similarity_values,
            25,
        )
    ),

    "p50": float(
        np.percentile(
            selected_similarity_values,
            50,
        )
    ),

    "p75": float(
        np.percentile(
            selected_similarity_values,
            75,
        )
    ),

    "p90": float(
        np.percentile(
            selected_similarity_values,
            90,
        )
    ),

    "max": float(
        selected_similarity_values.max()
    ),
}


# ---------------------------------------------------------
# Serialize immutable Candidate Selection artifact.
# ---------------------------------------------------------

semantic_selected_candidates_text = "".join(
    json.dumps(
        item,
        ensure_ascii=False,
        sort_keys=True,
    )
    + "\n"
    for item in (
        semantic_selected_candidates
    )
)


semantic_selected_candidates_sha256 = (
    hashlib.sha256(
        semantic_selected_candidates_text.encode(
            "utf-8"
        )
    ).hexdigest()
)


semantic_selection_report = {
    "report_version": (
        "claim_semantic_candidate_selection_report_v1"
    ),

    "selection_version": (
        SEMANTIC_SELECTION_VERSION
    ),

    "source_neighbor_candidates": (
        semantic_neighbor_candidates_path.name
    ),

    "source_neighbor_candidates_sha256": (
        semantic_neighbor_candidates_sha256
    ),

    "input_neighbor_pairs": (
        len(
            selection_candidates
        )
    ),

    "selection_policy": {
        "primary_similarity_threshold": (
            SEMANTIC_PRIMARY_THRESHOLD
        ),

        "mutual_recovery_similarity_threshold": (
            SEMANTIC_MUTUAL_RECOVERY_THRESHOLD
        ),

        "mutual_recovery_requires_mutual_top_k": (
            True
        ),
    },

    "selected_pairs": (
        len(
            semantic_selected_candidates
        )
    ),

    "selection_channel_counts": dict(
        sorted(
            selection_channel_counts.items()
        )
    ),

    "similarity_band_counts": dict(
        sorted(
            selection_band_counts.items()
        )
    ),

    "selected_mutual_top_k_pairs": (
        mutual_selected_count
    ),

    "selected_cross_source_pairs": (
        cross_source_selected_count
    ),

    "selected_similarity_stats": {
        key: round(
            value,
            8,
        )
        for (
            key,
            value,
        ) in (
            selected_similarity_stats.items()
        )
    },

    "semantic_relation_decisions_made": 0,

    "llm_calls": 0,

    "automatic_merge_applied": False,

    "claim_ids_assigned": False,

    "next_stage": (
        "SEMANTIC_RELATION_ADJUDICATION"
    ),

    "artifact": {
        "selected_candidates": (
            semantic_selected_candidates_path.name
        ),

        "selected_candidates_sha256": (
            semantic_selected_candidates_sha256
        ),
    },
}


semantic_selection_report_text = (
    json.dumps(
        semantic_selection_report,
        ensure_ascii=False,
        indent=2,
        sort_keys=True,
    )
    + "\n"
)


semantic_selected_candidates_write_status = (
    publish_frozen_artifact(
        semantic_selected_candidates_path,
        semantic_selected_candidates_text,
    )
)


semantic_selection_report_write_status = (
    publish_frozen_artifact(
        semantic_selection_report_path,
        semantic_selection_report_text,
    )
)


# ---------------------------------------------------------
# Independent disk revalidation.
# ---------------------------------------------------------

assert (
    load_jsonl(
        semantic_selected_candidates_path
    )
    == semantic_selected_candidates
)


assert (
    load_json(
        semantic_selection_report_path
    )
    == semantic_selection_report
)


assert (
    file_sha256(
        semantic_selected_candidates_path
    )
    == semantic_selected_candidates_sha256
)


semantic_candidate_selection_status = (
    "PASS"
)

semantic_relation_adjudication_readiness = (
    "READY_FOR_SEMANTIC_RELATION_ADJUDICATION"
)


# ---------------------------------------------------------
# Result.
# ---------------------------------------------------------

print(
    "=== NB12i SEMANTIC CANDIDATE SELECTION ==="
)

print()

print(
    "Input neighbor pairs:",
    len(
        selection_candidates
    ),
)

print(
    "Selected pairs:",
    len(
        semantic_selected_candidates
    ),
)


print()

print(
    "Selection channels:",
    dict(
        sorted(
            selection_channel_counts.items()
        )
    ),
)

print(
    "Similarity bands:",
    dict(
        sorted(
            selection_band_counts.items()
        )
    ),
)

print(
    "Selected Mutual Top-K pairs:",
    mutual_selected_count,
)

print(
    "Selected cross-source pairs:",
    cross_source_selected_count,
)


print()

print(
    "Selected similarity stats:"
)

for key in [
    "min",
    "p25",
    "p50",
    "p75",
    "p90",
    "max",
]:

    print(
        " ",
        key,
        "=",
        round(
            selected_similarity_stats[
                key
            ],
            6,
        ),
    )


print()

print(
    "LLM calls performed:",
    0,
)

print(
    "Semantic relation decisions made:",
    0,
)

print(
    "Automatic merge applied:",
    False,
)

print(
    "Claim IDs assigned:",
    False,
)


print()

print(
    "Selected Candidate Artifact:",
    semantic_selected_candidates_path.relative_to(
        project_root
    ),
)

print(
    "Selected Candidate SHA-256:",
    semantic_selected_candidates_sha256,
)

print(
    "Write status:",
    semantic_selected_candidates_write_status,
)


print()

print(
    "Next stage:",
    "SEMANTIC_RELATION_ADJUDICATION",
)

print(
    "Adjudication readiness:",
    semantic_relation_adjudication_readiness,
)


print()

print(
    "CODE CELL 23 GATE:",
    semantic_candidate_selection_status,
)
```

<div style="direction: rtl; white-space: normal; line-height: 1.7;">

<h2>Pilot داوری Semantic Relation</h2>

<p>
Candidate Selection با موفقیت پایان یافته و ۳۱۷۳ Pair برای بررسی معنایی انتخاب شده‌اند.
پیش از اجرای کامل، یک Pilot کوچک و نماینده از تمام بازه‌های Similarity اجرا می‌شود
تا قرارداد داوری Relation و رفتار مدل روی Duplicate، Subsumption، Partial Overlap و Distinct بررسی شود.
</p>

<ul>
<li>مدل همان <code>Qwen3-30B-A3B</code> باقی می‌ماند.</li>
<li>هیچ Claim حذف یا Merge نمی‌شود.</li>
<li>خروجی Pilot وارد Registry نهایی نمی‌شود.</li>
<li>نمونه‌ها از تمام Similarity Bandها و Pairهای Cross-Source انتخاب می‌شوند.</li>
<li>پس از PASS این Pilot، همان قرارداد برای کل ۳۱۷۳ Pair اجرا خواهد شد.</li>
</ul>

</div>

```python
# Code Cell 24

assert (
    semantic_candidate_selection_status == "PASS"
), (
    "Code Cell 23 must PASS before "
    "Semantic Relation Pilot."
)

assert (
    semantic_relation_adjudication_readiness
    == "READY_FOR_SEMANTIC_RELATION_ADJUDICATION"
)


# =========================================================
# Configuration
# =========================================================

RELATION_PILOT_VERSION = (
    "claim_semantic_relation_pilot_v1"
)

RELATION_PROMPT_VERSION = (
    "claim_semantic_relation_prompt_v1"
)

RELATION_PILOT_BATCH_SIZE = 4
RELATION_PILOT_MAX_ATTEMPTS = 3
RELATION_PILOT_PER_BAND = 8
RELATION_PILOT_EXTRA_CROSS_SOURCE = 8


RELATION_DECISIONS = {
    "SEMANTIC_DUPLICATE",
    "SUBSUMES",
    "PARTIAL_OVERLAP",
    "DISTINCT",
}

SUBSUMPTION_DIRECTIONS = {
    "A_SUBSUMES_B",
    "B_SUBSUMES_A",
}


relation_pilot_path = (
    registry_artifact_root
    / "claim_semantic_relation_pilot_v1.jsonl"
)

relation_pilot_report_path = (
    registry_artifact_root
    / "claim_semantic_relation_pilot_report_v1.json"
)


# =========================================================
# Independently reload selected Candidate Queue
# =========================================================

relation_candidates = load_jsonl(
    semantic_selected_candidates_path
)

relation_selection_report = load_json(
    semantic_selection_report_path
)


assert (
    file_sha256(
        semantic_selected_candidates_path
    )
    == semantic_selected_candidates_sha256
), (
    "Selected Semantic Candidate artifact changed."
)


assert (
    len(relation_candidates)
    == relation_selection_report[
        "selected_pairs"
    ]
)


assert (
    relation_selection_report[
        "semantic_relation_decisions_made"
    ]
    == 0
)


# =========================================================
# Deterministic stratified Pilot selection
# =========================================================

candidates_by_band = defaultdict(list)

for candidate in relation_candidates:
    candidates_by_band[
        candidate["similarity_band"]
    ].append(candidate)


expected_similarity_bands = [
    "0_70_TO_0_75_MUTUAL",
    "0_75_TO_0_80",
    "0_80_TO_0_85",
    "0_85_TO_0_90",
    "GE_0_90",
]


for band in expected_similarity_bands:

    assert candidates_by_band[band], (
        f"No candidates found for band: {band}"
    )


def deterministic_relation_sample(
    records,
    count,
):

    ordered = sorted(
        records,
        key=lambda item: hashlib.sha256(
            item[
                "candidate_pair_id"
            ].encode(
                "utf-8"
            )
        ).hexdigest(),
    )

    return ordered[
        :min(
            count,
            len(ordered),
        )
    ]


relation_pilot_by_id = {}


for band in expected_similarity_bands:

    selected = deterministic_relation_sample(
        candidates_by_band[band],
        RELATION_PILOT_PER_BAND,
    )

    for candidate in selected:

        relation_pilot_by_id[
            candidate["candidate_pair_id"]
        ] = candidate


cross_source_candidates = [
    candidate
    for candidate in relation_candidates
    if (
        set(candidate["source_types_a"])
        != set(candidate["source_types_b"])
    )
]


extra_cross_source = (
    deterministic_relation_sample(
        cross_source_candidates,
        RELATION_PILOT_EXTRA_CROSS_SOURCE,
    )
)


for candidate in extra_cross_source:

    relation_pilot_by_id[
        candidate["candidate_pair_id"]
    ] = candidate


relation_pilot_candidates = sorted(
    relation_pilot_by_id.values(),
    key=lambda item: (
        -item["cosine_similarity"],
        item["candidate_pair_id"],
    ),
)


assert (
    40
    <= len(relation_pilot_candidates)
    <= 48
)


pilot_candidate_by_id = {
    item["candidate_pair_id"]: item
    for item in relation_pilot_candidates
}


# =========================================================
# Relation Judge Prompt
# =========================================================

def build_relation_pilot_prompt(
    batch_candidates,
    retry_feedback=None,
):

    system_message = """
You are a strict semantic relation adjudicator for a
Persian evidence-grounded financial knowledge registry.

You receive pairs of already-validated Persian Claims.

For each pair, classify ONLY the semantic relationship
between Claim A and Claim B.

Allowed decisions:

1. SEMANTIC_DUPLICATE
   Both Claims express the same proposition.
   Wording may differ, but neither adds a meaningful
   condition, scope, quantity, cause, exception,
   attribution, or additional proposition.

2. SUBSUMES
   One Claim fully contains the meaning of the other
   and adds meaningful information.

   For this decision ONLY, use:
   subsumption_direction = "A_SUBSUMES_B"
   or
   subsumption_direction = "B_SUBSUMES_A"

3. PARTIAL_OVERLAP
   The Claims share meaningful factual content,
   but each contains information not fully contained
   in the other, or their scopes overlap only partly.

4. DISTINCT
   The Claims are related by topic or vocabulary,
   but do not satisfy the definitions above.

Important rules:

- Similar wording alone is NOT enough for duplicate.
- Same topic is NOT enough for overlap.
- Different numbers, conditions, populations,
  time periods, modalities, causal directions,
  perspectives, or scopes may make Claims distinct.
- Do not infer unstated information.
- Judge propositions exactly as written.
- Never rewrite the Claims.
- Never merge Claims.

STRICT OUTPUT CONTRACT:

For SEMANTIC_DUPLICATE:
  decision = "SEMANTIC_DUPLICATE"
  subsumption_direction = null

For SUBSUMES:
  decision = "SUBSUMES"
  subsumption_direction =
    "A_SUBSUMES_B" or "B_SUBSUMES_A"

For PARTIAL_OVERLAP:
  decision = "PARTIAL_OVERLAP"
  subsumption_direction = null

For DISTINCT:
  decision = "DISTINCT"
  subsumption_direction = null

Output JSON only.
""".strip()


    batch_payload = []

    for candidate in batch_candidates:

        batch_payload.append({
            "candidate_pair_id": (
                candidate[
                    "candidate_pair_id"
                ]
            ),

            "claim_a": (
                candidate[
                    "claim_text_a"
                ]
            ),

            "claim_b": (
                candidate[
                    "claim_text_b"
                ]
            ),

            "claim_types_a": (
                candidate[
                    "claim_types_a"
                ]
            ),

            "claim_types_b": (
                candidate[
                    "claim_types_b"
                ]
            ),

            "cosine_similarity": (
                candidate[
                    "cosine_similarity"
                ]
            ),
        })


    required_schema = {
        "judgments": [
            {
                "candidate_pair_id": (
                    "exact input ID"
                ),

                "decision": (
                    "SEMANTIC_DUPLICATE | "
                    "SUBSUMES | "
                    "PARTIAL_OVERLAP | "
                    "DISTINCT"
                ),

                "subsumption_direction": (
                    "A_SUBSUMES_B | "
                    "B_SUBSUMES_A | null"
                ),

                "brief_reason_fa": (
                    "short Persian explanation"
                ),
            }
        ]
    }


    user_payload = {
        "instruction": (
            "Classify every pair exactly once."
        ),

        "required_schema": (
            required_schema
        ),

        "pairs": (
            batch_payload
        ),
    }


    if retry_feedback:

        user_payload[
            "retry_feedback"
        ] = (
            "Your previous response was invalid. "
            "Correct ONLY the structural or contract "
            "problem described here: "
            + retry_feedback
        )


    user_message = json.dumps(
        user_payload,
        ensure_ascii=False,
        indent=2,
    )


    return (
        system_message,
        user_message,
    )


# =========================================================
# Strict payload validator + harmless schema normalization
# =========================================================

relation_pilot_normalization_counts = Counter()


def validate_relation_pilot_payload(
    batch_candidates,
    payload,
):

    if not isinstance(
        payload,
        dict,
    ):

        raise ValueError(
            "Relation response must be an object."
        )


    judgments = payload.get(
        "judgments"
    )


    if not isinstance(
        judgments,
        list,
    ):

        raise ValueError(
            "Relation response requires judgments list."
        )


    expected_ids = [
        candidate[
            "candidate_pair_id"
        ]
        for candidate in batch_candidates
    ]


    if (
        len(judgments)
        != len(expected_ids)
    ):

        raise ValueError(
            "Relation judgment count does not "
            "match batch."
        )


    seen_ids = set()
    canonical = []


    for judgment in judgments:

        if not isinstance(
            judgment,
            dict,
        ):

            raise ValueError(
                "Each Relation judgment "
                "must be an object."
            )


        candidate_pair_id = (
            judgment.get(
                "candidate_pair_id"
            )
        )


        if (
            candidate_pair_id
            not in expected_ids
        ):

            raise ValueError(
                "Unknown candidate_pair_id "
                "in Relation response."
            )


        if (
            candidate_pair_id
            in seen_ids
        ):

            raise ValueError(
                "Duplicate candidate_pair_id "
                "in Relation response."
            )


        seen_ids.add(
            candidate_pair_id
        )


        # -------------------------------------------------
        # Read raw model fields
        # -------------------------------------------------

        decision = judgment.get(
            "decision"
        )

        direction = judgment.get(
            "subsumption_direction"
        )


        if isinstance(
            decision,
            str,
        ):
            decision = decision.strip()


        if isinstance(
            direction,
            str,
        ):
            direction = direction.strip()


        # -------------------------------------------------
        # NORMALIZATION 1
        #
        # Qwen sometimes emits:
        #
        # decision = "A_SUBSUMES_B"
        #
        # instead of:
        #
        # decision = "SUBSUMES"
        # direction = "A_SUBSUMES_B"
        #
        # Semantically identical -> canonicalize.
        # -------------------------------------------------

        if (
            decision
            in SUBSUMPTION_DIRECTIONS
        ):

            direction = decision
            decision = "SUBSUMES"

            relation_pilot_normalization_counts[
                "DIRECTION_EMITTED_AS_DECISION"
            ] += 1


        # -------------------------------------------------
        # Validate canonical decision
        # -------------------------------------------------

        if (
            decision
            not in RELATION_DECISIONS
        ):

            raise ValueError(
                "Invalid Relation decision: "
                f"{decision}"
            )


        # -------------------------------------------------
        # SUBSUMES requires valid direction
        # -------------------------------------------------

        if (
            decision
            == "SUBSUMES"
        ):

            if (
                direction
                not in SUBSUMPTION_DIRECTIONS
            ):

                raise ValueError(
                    "SUBSUMES requires a valid "
                    "subsumption_direction."
                )


        # -------------------------------------------------
        # NORMALIZATION 2
        #
        # Qwen sometimes provides a stray direction for:
        # DUPLICATE / PARTIAL_OVERLAP / DISTINCT.
        #
        # Direction is semantically meaningless there,
        # therefore canonicalize to null.
        # -------------------------------------------------

        else:

            if (
                direction
                is not None
            ):

                relation_pilot_normalization_counts[
                    "NON_SUBSUMES_DIRECTION_CLEARED"
                ] += 1

            direction = None


        brief_reason_fa = judgment.get(
            "brief_reason_fa"
        )


        if not (
            isinstance(
                brief_reason_fa,
                str,
            )
            and brief_reason_fa.strip()
        ):

            raise ValueError(
                "brief_reason_fa must be "
                "a non-empty string."
            )


        canonical.append({
            "candidate_pair_id": (
                candidate_pair_id
            ),

            "decision": (
                decision
            ),

            "subsumption_direction": (
                direction
            ),

            "brief_reason_fa": (
                brief_reason_fa.strip()
            ),
        })


    if (
        seen_ids
        != set(expected_ids)
    ):

        raise ValueError(
            "Relation response IDs do not "
            "exactly match batch."
        )


    canonical_by_id = {
        item[
            "candidate_pair_id"
        ]: item
        for item in canonical
    }


    return [
        canonical_by_id[
            candidate_pair_id
        ]
        for candidate_pair_id
        in expected_ids
    ]


# =========================================================
# Rerun safety
#
# If BOTH frozen Pilot artifacts already exist and validate,
# reuse them without making new API calls.
# =========================================================

pilot_artifact_exists = (
    relation_pilot_path.exists()
)

pilot_report_exists = (
    relation_pilot_report_path.exists()
)


assert (
    pilot_artifact_exists
    == pilot_report_exists
), (
    "Pilot artifact/report existence mismatch. "
    "Expected both or neither."
)


relation_pilot_reused_existing = False

relation_pilot_api_calls = 0
relation_pilot_retry_events = 0


if (
    pilot_artifact_exists
    and pilot_report_exists
):

    existing_relation_pilot_report = (
        load_json(
            relation_pilot_report_path
        )
    )


    assert (
        existing_relation_pilot_report[
            "pilot_version"
        ]
        == RELATION_PILOT_VERSION
    )


    assert (
        existing_relation_pilot_report[
            "relation_prompt_version"
        ]
        == RELATION_PROMPT_VERSION
    )


    assert (
        existing_relation_pilot_report[
            "semantic_model"
        ]
        == pilot_chat_model
    )


    assert (
        existing_relation_pilot_report[
            "source_selected_candidates_sha256"
        ]
        == semantic_selected_candidates_sha256
    )


    existing_pilot_sha256 = (
        file_sha256(
            relation_pilot_path
        )
    )


    assert (
        existing_pilot_sha256
        == existing_relation_pilot_report[
            "artifact"
        ][
            "pilot_sha256"
        ]
    )


    relation_pilot_records = (
        load_jsonl(
            relation_pilot_path
        )
    )


    assert (
        len(relation_pilot_records)
        == len(
            relation_pilot_candidates
        )
    )


    assert (
        {
            item[
                "candidate_pair_id"
            ]
            for item in relation_pilot_records
        }
        == {
            item[
                "candidate_pair_id"
            ]
            for item in relation_pilot_candidates
        }
    )


    relation_pilot_api_calls = 0
    relation_pilot_retry_events = 0

    relation_pilot_reused_existing = True


    relation_pilot_sha256 = (
        existing_pilot_sha256
    )

    relation_pilot_write_status = (
        "REUSED_EXISTING_IDENTICAL"
    )

    relation_pilot_report_write_status = (
        "REUSED_EXISTING_IDENTICAL"
    )


# =========================================================
# Otherwise run the Pilot
# =========================================================

else:

    relation_pilot_batches = [
        relation_pilot_candidates[
            start:
            start
            + RELATION_PILOT_BATCH_SIZE
        ]
        for start in range(
            0,
            len(
                relation_pilot_candidates
            ),
            RELATION_PILOT_BATCH_SIZE,
        )
    ]


    relation_pilot_judgments = []


    print(
        "=== NB12i SEMANTIC RELATION PILOT ==="
    )

    print()

    print(
        "Prompt version:",
        RELATION_PROMPT_VERSION,
    )

    print(
        "Model:",
        pilot_chat_model,
    )

    print(
        "Pilot Candidate pairs:",
        len(
            relation_pilot_candidates
        ),
    )

    print(
        "Pilot batches:",
        len(
            relation_pilot_batches
        ),
    )

    print(
        "Batch size:",
        RELATION_PILOT_BATCH_SIZE,
    )

    print()


    for (
        batch_number,
        batch_candidates,
    ) in enumerate(
        relation_pilot_batches,
        start=1,
    ):

        validated_judgments = None
        last_error = None


        for attempt in range(
            1,
            RELATION_PILOT_MAX_ATTEMPTS
            + 1,
        ):

            (
                system_message,
                user_message,
            ) = build_relation_pilot_prompt(
                batch_candidates,
                retry_feedback=(
                    last_error
                ),
            )


            try:

                relation_pilot_api_calls += 1


                content = (
                    call_semantic_v3_judge(
                        system_message,
                        user_message,
                    )
                )


                parsed_payload = (
                    parse_model_json(
                        content
                    )
                )


                validated_judgments = (
                    validate_relation_pilot_payload(
                        batch_candidates,
                        parsed_payload,
                    )
                )


                break


            except Exception as exc:

                last_error = (
                    f"{type(exc).__name__}: "
                    f"{exc}"
                )


                relation_pilot_retry_events += 1


                print(
                    "  batch",
                    batch_number,
                    "attempt",
                    attempt,
                    "INVALID |",
                    last_error,
                )


        assert (
            validated_judgments
            is not None
        ), (
            "Relation Pilot batch failed after "
            f"{RELATION_PILOT_MAX_ATTEMPTS} "
            "attempts: "
            f"batch {batch_number}"
        )


        relation_pilot_judgments.extend(
            validated_judgments
        )


        print(
            "[",
            batch_number,
            "/",
            len(
                relation_pilot_batches
            ),
            "] VALID | pairs:",
            len(
                batch_candidates
            ),
        )


    # =====================================================
    # Join Candidate metadata with judgments
    # =====================================================

    relation_pilot_records = []


    for judgment in relation_pilot_judgments:

        candidate = (
            pilot_candidate_by_id[
                judgment[
                    "candidate_pair_id"
                ]
            ]
        )


        relation_pilot_records.append({
            "candidate_pair_id": (
                judgment[
                    "candidate_pair_id"
                ]
            ),

            "claim_text_a": (
                candidate[
                    "claim_text_a"
                ]
            ),

            "claim_text_b": (
                candidate[
                    "claim_text_b"
                ]
            ),

            "cosine_similarity": (
                candidate[
                    "cosine_similarity"
                ]
            ),

            "similarity_band": (
                candidate[
                    "similarity_band"
                ]
            ),

            "mutual_top_k": (
                candidate[
                    "mutual_top_k"
                ]
            ),

            "source_types_a": (
                candidate[
                    "source_types_a"
                ]
            ),

            "source_types_b": (
                candidate[
                    "source_types_b"
                ]
            ),

            "decision": (
                judgment[
                    "decision"
                ]
            ),

            "subsumption_direction": (
                judgment[
                    "subsumption_direction"
                ]
            ),

            "brief_reason_fa": (
                judgment[
                    "brief_reason_fa"
                ]
            ),

            "semantic_model": (
                pilot_chat_model
            ),

            "relation_prompt_version": (
                RELATION_PROMPT_VERSION
            ),

            "pilot_version": (
                RELATION_PILOT_VERSION
            ),
        })


    relation_pilot_records = sorted(
        relation_pilot_records,
        key=lambda item: (
            -item[
                "cosine_similarity"
            ],
            item[
                "candidate_pair_id"
            ],
        ),
    )


    assert (
        len(relation_pilot_records)
        == len(
            relation_pilot_candidates
        )
    )


    assert (
        len({
            item[
                "candidate_pair_id"
            ]
            for item in relation_pilot_records
        })
        == len(
            relation_pilot_records
        )
    )


    # =====================================================
    # Diagnostics before freezing
    # =====================================================

    relation_pilot_decision_counts = Counter(
        item["decision"]
        for item in relation_pilot_records
    )


    relation_pilot_by_band = defaultdict(
        Counter
    )


    for item in relation_pilot_records:

        relation_pilot_by_band[
            item[
                "similarity_band"
            ]
        ][
            item[
                "decision"
            ]
        ] += 1


    relation_pilot_cross_source_count = sum(
        1
        for item in relation_pilot_records
        if (
            set(
                item["source_types_a"]
            )
            != set(
                item["source_types_b"]
            )
        )
    )


    # =====================================================
    # Freeze Pilot artifact
    # =====================================================

    relation_pilot_text = "".join(
        json.dumps(
            item,
            ensure_ascii=False,
            sort_keys=True,
        )
        + "\n"
        for item in relation_pilot_records
    )


    relation_pilot_sha256 = (
        hashlib.sha256(
            relation_pilot_text.encode(
                "utf-8"
            )
        ).hexdigest()
    )


    relation_pilot_report = {
        "report_version": (
            "claim_semantic_relation_pilot_report_v1"
        ),

        "pilot_version": (
            RELATION_PILOT_VERSION
        ),

        "relation_prompt_version": (
            RELATION_PROMPT_VERSION
        ),

        "semantic_model": (
            pilot_chat_model
        ),

        "source_selected_candidates": (
            semantic_selected_candidates_path.name
        ),

        "source_selected_candidates_sha256": (
            semantic_selected_candidates_sha256
        ),

        "pilot_pairs": (
            len(
                relation_pilot_records
            )
        ),

        "pilot_batches": (
            len(
                relation_pilot_batches
            )
        ),

        "api_calls": (
            relation_pilot_api_calls
        ),

        "retry_events": (
            relation_pilot_retry_events
        ),

        "schema_normalizations": dict(
            sorted(
                relation_pilot_normalization_counts.items()
            )
        ),

        "decision_counts": dict(
            sorted(
                relation_pilot_decision_counts.items()
            )
        ),

        "cross_source_pairs": (
            relation_pilot_cross_source_count
        ),

        "decisions_by_similarity_band": {
            band: dict(
                sorted(
                    counts.items()
                )
            )
            for (
                band,
                counts,
            ) in sorted(
                relation_pilot_by_band.items()
            )
        },

        "automatic_merge_applied": False,

        "claim_ids_assigned": False,

        "full_relation_adjudication": (
            "NOT_STARTED"
        ),

        "artifact": {
            "pilot": (
                relation_pilot_path.name
            ),

            "pilot_sha256": (
                relation_pilot_sha256
            ),
        },
    }


    relation_pilot_report_text = (
        json.dumps(
            relation_pilot_report,
            ensure_ascii=False,
            indent=2,
            sort_keys=True,
        )
        + "\n"
    )


    relation_pilot_write_status = (
        publish_frozen_artifact(
            relation_pilot_path,
            relation_pilot_text,
        )
    )


    relation_pilot_report_write_status = (
        publish_frozen_artifact(
            relation_pilot_report_path,
            relation_pilot_report_text,
        )
    )


# =========================================================
# Reload frozen Pilot for common validation/output path
# =========================================================

relation_pilot_records = load_jsonl(
    relation_pilot_path
)

relation_pilot_report = load_json(
    relation_pilot_report_path
)


relation_pilot_sha256 = (
    file_sha256(
        relation_pilot_path
    )
)


assert (
    relation_pilot_sha256
    == relation_pilot_report[
        "artifact"
    ][
        "pilot_sha256"
    ]
)


assert (
    relation_pilot_report[
        "source_selected_candidates_sha256"
    ]
    == semantic_selected_candidates_sha256
)


assert (
    len(relation_pilot_records)
    == len(
        relation_pilot_candidates
    )
)


assert (
    {
        item[
            "candidate_pair_id"
        ]
        for item in relation_pilot_records
    }
    == {
        item[
            "candidate_pair_id"
        ]
        for item in relation_pilot_candidates
    }
)


# =========================================================
# Validate canonical Relation schema on frozen records
# =========================================================

for item in relation_pilot_records:

    assert (
        item["decision"]
        in RELATION_DECISIONS
    )


    if (
        item["decision"]
        == "SUBSUMES"
    ):

        assert (
            item[
                "subsumption_direction"
            ]
            in SUBSUMPTION_DIRECTIONS
        )


    else:

        assert (
            item[
                "subsumption_direction"
            ]
            is None
        )


# =========================================================
# Final diagnostics from frozen artifact
# =========================================================

relation_pilot_decision_counts = Counter(
    item["decision"]
    for item in relation_pilot_records
)


relation_pilot_by_band = defaultdict(
    Counter
)


for item in relation_pilot_records:

    relation_pilot_by_band[
        item[
            "similarity_band"
        ]
    ][
        item[
            "decision"
        ]
    ] += 1


relation_pilot_cross_source_count = sum(
    1
    for item in relation_pilot_records
    if (
        set(
            item[
                "source_types_a"
            ]
        )
        != set(
            item[
                "source_types_b"
            ]
        )
    )
)


# =========================================================
# Gate
# =========================================================

semantic_relation_pilot_status = (
    "PASS"
)

semantic_relation_full_readiness = (
    "READY_FOR_FULL_RELATION_ADJUDICATION"
)


# =========================================================
# Result
# =========================================================

if relation_pilot_reused_existing:

    print(
        "=== NB12i SEMANTIC RELATION PILOT ==="
    )

    print()

    print(
        "Existing frozen Pilot reused:",
        True,
    )

    print(
        "API calls THIS run:",
        0,
    )

    print()


print(
    "=== SEMANTIC RELATION PILOT RESULT ==="
)

print()

print(
    "Prompt version:",
    RELATION_PROMPT_VERSION,
)

print(
    "Model:",
    pilot_chat_model,
)

print(
    "Pilot pairs:",
    len(
        relation_pilot_records
    ),
)

print(
    "API calls THIS run:",
    relation_pilot_api_calls,
)

print(
    "Retry events THIS run:",
    relation_pilot_retry_events,
)

print(
    "Pilot reused existing:",
    relation_pilot_reused_existing,
)

print(
    "Cross-source Pilot pairs:",
    relation_pilot_cross_source_count,
)


print()

print(
    "Relation decisions:",
    dict(
        sorted(
            relation_pilot_decision_counts.items()
        )
    ),
)


print()

print(
    "Schema normalizations:"
)

print(
    relation_pilot_report.get(
        "schema_normalizations",
        {},
    )
)


print()

print(
    "Decisions by similarity band:"
)


for band in expected_similarity_bands:

    print(
        " ",
        band,
        ":",
        dict(
            sorted(
                relation_pilot_by_band[
                    band
                ].items()
            )
        ),
    )


print()

print(
    "Representative examples:"
)


for decision in [
    "SEMANTIC_DUPLICATE",
    "SUBSUMES",
    "PARTIAL_OVERLAP",
    "DISTINCT",
]:

    matching = [
        item
        for item in relation_pilot_records
        if (
            item["decision"]
            == decision
        )
    ]


    print()

    print(
        "---",
        decision,
        "---",
    )


    if not matching:

        print(
            "  NONE"
        )

        continue


    for item in matching[:2]:

        print(
            "  similarity:",
            item[
                "cosine_similarity"
            ],
        )

        print(
            "  A:",
            item[
                "claim_text_a"
            ][
                :300
            ],
        )

        print(
            "  B:",
            item[
                "claim_text_b"
            ][
                :300
            ],
        )

        print(
            "  direction:",
            item[
                "subsumption_direction"
            ],
        )

        print(
            "  reason:",
            item[
                "brief_reason_fa"
            ],
        )

        print()


print(
    "Automatic merge applied:",
    False,
)

print(
    "Full adjudication started:",
    False,
)


print()

print(
    "Pilot Artifact:",
    relation_pilot_path.relative_to(
        project_root
    ),
)

print(
    "Pilot SHA-256:",
    relation_pilot_sha256,
)

print(
    "Pilot write status:",
    relation_pilot_write_status,
)

print(
    "Report write status:",
    relation_pilot_report_write_status,
)


print()

print(
    "Full adjudication readiness:",
    semantic_relation_full_readiness,
)


print()

print(
    "CODE CELL 24 GATE:",
    semantic_relation_pilot_status,
)
```

```python
# Code Cell 25

# Temporary development imports.
# Move these to Code Cell 01 before final Restart -> Run All.
from concurrent.futures import (
    ThreadPoolExecutor,
    as_completed,
)
import threading
import time


assert (
    semantic_relation_pilot_status
    == "PASS"
), (
    "Code Cell 24 must PASS before "
    "Full Semantic Relation Adjudication."
)

assert (
    semantic_relation_full_readiness
    == "READY_FOR_FULL_RELATION_ADJUDICATION"
)


# =========================================================
# Configuration
# =========================================================

FULL_RELATION_VERSION = (
    "claim_semantic_relation_full_v1"
)

FULL_RELATION_CACHE_VERSION = (
    "claim_semantic_relation_cache_v1"
)

FULL_RELATION_BATCH_SIZE = 4

FULL_RELATION_MAX_ATTEMPTS = 4

FULL_RELATION_PROVIDER_RPM = 30

FULL_RELATION_INTERNAL_RPM = 24

FULL_RELATION_WORKERS = 6


assert (
    FULL_RELATION_INTERNAL_RPM
    <= FULL_RELATION_PROVIDER_RPM
)


full_relation_cache_root = (
    registry_artifact_root
    / "claim_semantic_relation_cache_v1"
)

full_relation_artifact_path = (
    registry_artifact_root
    / "claim_semantic_relation_full_adjudication_v1.jsonl"
)

full_relation_report_path = (
    registry_artifact_root
    / "claim_semantic_relation_full_report_v1.json"
)


full_relation_cache_root.mkdir(
    parents=True,
    exist_ok=True,
)


# =========================================================
# Reload and validate frozen inputs
# =========================================================

full_relation_candidates = load_jsonl(
    semantic_selected_candidates_path
)

full_relation_selection_report = load_json(
    semantic_selection_report_path
)

full_relation_pilot_records = load_jsonl(
    relation_pilot_path
)

full_relation_pilot_report = load_json(
    relation_pilot_report_path
)


assert (
    file_sha256(
        semantic_selected_candidates_path
    )
    == semantic_selected_candidates_sha256
)


assert (
    len(
        full_relation_candidates
    )
    == full_relation_selection_report[
        "selected_pairs"
    ]
)


assert (
    file_sha256(
        relation_pilot_path
    )
    == full_relation_pilot_report[
        "artifact"
    ][
        "pilot_sha256"
    ]
)


assert (
    full_relation_pilot_report[
        "source_selected_candidates_sha256"
    ]
    == semantic_selected_candidates_sha256
)


assert (
    full_relation_pilot_report[
        "relation_prompt_version"
    ]
    == RELATION_PROMPT_VERSION
)


assert (
    full_relation_pilot_report[
        "semantic_model"
    ]
    == pilot_chat_model
)


# =========================================================
# Candidate and Pilot indexes
# =========================================================

full_candidate_by_id = {
    item[
        "candidate_pair_id"
    ]: item
    for item in (
        full_relation_candidates
    )
}


assert (
    len(
        full_candidate_by_id
    )
    == len(
        full_relation_candidates
    )
)


pilot_relation_by_id = {
    item[
        "candidate_pair_id"
    ]: item
    for item in (
        full_relation_pilot_records
    )
}


assert (
    len(
        pilot_relation_by_id
    )
    == len(
        full_relation_pilot_records
    )
)


assert (
    set(
        pilot_relation_by_id
    )
    <= set(
        full_candidate_by_id
    )
)


# =========================================================
# Validate canonical Pilot decisions
# =========================================================

for item in (
    full_relation_pilot_records
):

    assert (
        item[
            "decision"
        ]
        in RELATION_DECISIONS
    )


    if (
        item[
            "decision"
        ]
        == "SUBSUMES"
    ):

        assert (
            item[
                "subsumption_direction"
            ]
            in SUBSUMPTION_DIRECTIONS
        )


    else:

        assert (
            item[
                "subsumption_direction"
            ]
            is None
        )


# =========================================================
# Remaining candidates only
# =========================================================

full_api_candidates = [
    candidate
    for candidate in (
        full_relation_candidates
    )
    if (
        candidate[
            "candidate_pair_id"
        ]
        not in pilot_relation_by_id
    )
]


assert (
    len(
        full_api_candidates
    )
    + len(
        full_relation_pilot_records
    )
    == len(
        full_relation_candidates
    )
)


# =========================================================
# Deterministic batches
# =========================================================

full_relation_batches = [
    full_api_candidates[
        start:
        start
        + FULL_RELATION_BATCH_SIZE
    ]
    for start in range(
        0,
        len(
            full_api_candidates
        ),
        FULL_RELATION_BATCH_SIZE,
    )
]


assert all(
    1
    <= len(batch)
    <= FULL_RELATION_BATCH_SIZE
    for batch in (
        full_relation_batches
    )
)


def full_relation_batch_id(
    batch_candidates,
):

    material = (
        FULL_RELATION_VERSION
        + "\n"
        + RELATION_PROMPT_VERSION
        + "\n"
        + pilot_chat_model
        + "\n"
        + "\n".join(
            candidate[
                "candidate_pair_id"
            ]
            for candidate in (
                batch_candidates
            )
        )
    )


    digest = hashlib.sha256(
        material.encode(
            "utf-8"
        )
    ).hexdigest()


    return (
        "relation-batch:"
        + digest[:24]
    )


def full_relation_cache_path(
    batch_candidates,
):

    batch_id = full_relation_batch_id(
        batch_candidates
    )


    return (
        full_relation_cache_root
        / (
            batch_id.replace(
                ":",
                "_",
            )
            + ".json"
        )
    )


# =========================================================
# Full relation payload validator
#
# Same semantic contract as the successful Pilot.
# Harmless Qwen schema variants are normalized.
# =========================================================

def validate_full_relation_payload(
    batch_candidates,
    payload,
):

    if not isinstance(
        payload,
        dict,
    ):

        raise ValueError(
            "Relation response must be an object."
        )


    judgments = payload.get(
        "judgments"
    )


    if not isinstance(
        judgments,
        list,
    ):

        raise ValueError(
            "Relation response requires judgments list."
        )


    expected_ids = [
        candidate[
            "candidate_pair_id"
        ]
        for candidate in (
            batch_candidates
        )
    ]


    if (
        len(
            judgments
        )
        != len(
            expected_ids
        )
    ):

        raise ValueError(
            "Relation judgment count does not "
            "match batch."
        )


    seen_ids = set()

    canonical = []

    normalization_counts = Counter()


    for judgment in judgments:

        if not isinstance(
            judgment,
            dict,
        ):

            raise ValueError(
                "Each Relation judgment "
                "must be an object."
            )


        candidate_pair_id = (
            judgment.get(
                "candidate_pair_id"
            )
        )


        if (
            candidate_pair_id
            not in expected_ids
        ):

            raise ValueError(
                "Unknown candidate_pair_id "
                "in Relation response."
            )


        if (
            candidate_pair_id
            in seen_ids
        ):

            raise ValueError(
                "Duplicate candidate_pair_id "
                "in Relation response."
            )


        seen_ids.add(
            candidate_pair_id
        )


        decision = judgment.get(
            "decision"
        )

        direction = judgment.get(
            "subsumption_direction"
        )


        if isinstance(
            decision,
            str,
        ):
            decision = decision.strip()


        if isinstance(
            direction,
            str,
        ):
            direction = direction.strip()


        # -------------------------------------------------
        # Harmless schema normalization:
        #
        # decision = A_SUBSUMES_B
        # becomes
        # decision = SUBSUMES
        # direction = A_SUBSUMES_B
        # -------------------------------------------------

        if (
            decision
            in SUBSUMPTION_DIRECTIONS
        ):

            direction = decision

            decision = "SUBSUMES"

            normalization_counts[
                "DIRECTION_EMITTED_AS_DECISION"
            ] += 1


        if (
            decision
            not in RELATION_DECISIONS
        ):

            raise ValueError(
                "Invalid Relation decision: "
                f"{decision}"
            )


        if (
            decision
            == "SUBSUMES"
        ):

            if (
                direction
                not in SUBSUMPTION_DIRECTIONS
            ):

                raise ValueError(
                    "SUBSUMES requires a valid "
                    "subsumption_direction."
                )


        else:

            if (
                direction
                is not None
            ):

                normalization_counts[
                    "NON_SUBSUMES_DIRECTION_CLEARED"
                ] += 1


            direction = None


        brief_reason_fa = judgment.get(
            "brief_reason_fa"
        )


        if not (
            isinstance(
                brief_reason_fa,
                str,
            )
            and brief_reason_fa.strip()
        ):

            raise ValueError(
                "brief_reason_fa must be "
                "a non-empty string."
            )


        canonical.append({
            "candidate_pair_id": (
                candidate_pair_id
            ),

            "decision": (
                decision
            ),

            "subsumption_direction": (
                direction
            ),

            "brief_reason_fa": (
                brief_reason_fa.strip()
            ),
        })


    if (
        seen_ids
        != set(
            expected_ids
        )
    ):

        raise ValueError(
            "Relation response IDs do not "
            "exactly match batch."
        )


    canonical_by_id = {
        item[
            "candidate_pair_id"
        ]: item
        for item in canonical
    }


    canonical = [
        canonical_by_id[
            candidate_pair_id
        ]
        for candidate_pair_id
        in expected_ids
    ]


    return (
        canonical,
        dict(
            sorted(
                normalization_counts.items()
            )
        ),
    )


# =========================================================
# Cache validator
# =========================================================

def validate_full_relation_cache(
    batch_candidates,
    cache_payload,
):

    expected_batch_id = (
        full_relation_batch_id(
            batch_candidates
        )
    )


    expected_candidate_ids = [
        candidate[
            "candidate_pair_id"
        ]
        for candidate in (
            batch_candidates
        )
    ]


    assert (
        cache_payload[
            "cache_version"
        ]
        == FULL_RELATION_CACHE_VERSION
    )


    assert (
        cache_payload[
            "full_relation_version"
        ]
        == FULL_RELATION_VERSION
    )


    assert (
        cache_payload[
            "batch_id"
        ]
        == expected_batch_id
    )


    assert (
        cache_payload[
            "semantic_model"
        ]
        == pilot_chat_model
    )


    assert (
        cache_payload[
            "relation_prompt_version"
        ]
        == RELATION_PROMPT_VERSION
    )


    assert (
        cache_payload[
            "source_selected_candidates_sha256"
        ]
        == semantic_selected_candidates_sha256
    )


    assert (
        cache_payload[
            "candidate_pair_ids"
        ]
        == expected_candidate_ids
    )


    assert (
        isinstance(
            cache_payload[
                "attempt_count"
            ],
            int,
        )
        and cache_payload[
            "attempt_count"
        ]
        >= 1
    )


    assert (
        cache_payload[
            "retry_events"
        ]
        == (
            cache_payload[
                "attempt_count"
            ]
            - 1
        )
    )


    canonical_judgments = (
        cache_payload[
            "judgments"
        ]
    )


    assert (
        len(
            canonical_judgments
        )
        == len(
            expected_candidate_ids
        )
    )


    seen_ids = set()


    for judgment in canonical_judgments:

        candidate_pair_id = (
            judgment[
                "candidate_pair_id"
            ]
        )


        assert (
            candidate_pair_id
            in expected_candidate_ids
        )


        assert (
            candidate_pair_id
            not in seen_ids
        )


        seen_ids.add(
            candidate_pair_id
        )


        decision = (
            judgment[
                "decision"
            ]
        )


        direction = (
            judgment[
                "subsumption_direction"
            ]
        )


        assert (
            decision
            in RELATION_DECISIONS
        )


        if (
            decision
            == "SUBSUMES"
        ):

            assert (
                direction
                in SUBSUMPTION_DIRECTIONS
            )


        else:

            assert (
                direction
                is None
            )


        assert (
            isinstance(
                judgment[
                    "brief_reason_fa"
                ],
                str,
            )
            and judgment[
                "brief_reason_fa"
            ].strip()
        )


    assert (
        seen_ids
        == set(
            expected_candidate_ids
        )
    )


    normalization_counts = (
        cache_payload[
            "schema_normalizations"
        ]
    )


    assert isinstance(
        normalization_counts,
        dict,
    )


    for (
        key,
        value,
    ) in (
        normalization_counts.items()
    ):

        assert isinstance(
            key,
            str,
        )

        assert (
            isinstance(
                value,
                int,
            )
            and value >= 0
        )


    return cache_payload


# =========================================================
# Inspect existing caches before API work
# =========================================================

existing_full_relation_caches = 0

pending_full_relation_batches = []


for batch_candidates in (
    full_relation_batches
):

    cache_path = (
        full_relation_cache_path(
            batch_candidates
        )
    )


    if cache_path.exists():

        cache_payload = load_json(
            cache_path
        )


        validate_full_relation_cache(
            batch_candidates,
            cache_payload,
        )


        existing_full_relation_caches += 1


    else:

        pending_full_relation_batches.append(
            batch_candidates
        )


# =========================================================
# Progress header
# =========================================================

print(
    "=== NB12i FULL SEMANTIC RELATION ADJUDICATION ==="
)

print()

print(
    "Model:",
    pilot_chat_model,
)

print(
    "Prompt version:",
    RELATION_PROMPT_VERSION,
)

print(
    "Selected Candidate pairs:",
    len(
        full_relation_candidates
    ),
)

print(
    "Pilot judgments reused:",
    len(
        full_relation_pilot_records
    ),
)

print(
    "Pairs requiring Full adjudication:",
    len(
        full_api_candidates
    ),
)

print(
    "Full relation batches:",
    len(
        full_relation_batches
    ),
)

print(
    "Validated Full caches before run:",
    existing_full_relation_caches,
)

print(
    "Remaining Full batches:",
    len(
        pending_full_relation_batches
    ),
)

print()

print(
    "Workers:",
    FULL_RELATION_WORKERS,
)

print(
    "Provider Rate Limit:",
    FULL_RELATION_PROVIDER_RPM,
    "requests/minute",
)

print(
    "Internal Rate Limit:",
    FULL_RELATION_INTERNAL_RPM,
    "requests/minute",
)


if pending_full_relation_batches:

    print(
        "Approximate request-start floor:",
        round(
            len(
                pending_full_relation_batches
            )
            / float(
                FULL_RELATION_INTERNAL_RPM
            ),
            1,
        ),
        "minutes",
    )


print()


# =========================================================
# Thread-safe global rate limiter
# =========================================================

full_relation_rate_lock = (
    threading.Lock()
)

full_relation_print_lock = (
    threading.Lock()
)


full_relation_request_interval = (
    60.0
    / float(
        FULL_RELATION_INTERNAL_RPM
    )
)


full_relation_next_request_time = (
    time.monotonic()
)


def wait_for_full_relation_slot():

    global full_relation_next_request_time


    with full_relation_rate_lock:

        now = time.monotonic()


        wait_seconds = max(
            0.0,
            (
                full_relation_next_request_time
                - now
            ),
        )


        if wait_seconds > 0:

            time.sleep(
                wait_seconds
            )


        actual_start = (
            time.monotonic()
        )


        full_relation_next_request_time = (
            max(
                actual_start,
                full_relation_next_request_time,
            )
            + full_relation_request_interval
        )


# =========================================================
# One resume-safe Batch adjudication
# =========================================================

def adjudicate_full_relation_batch(
    batch_candidates,
):

    batch_id = (
        full_relation_batch_id(
            batch_candidates
        )
    )


    cache_path = (
        full_relation_cache_path(
            batch_candidates
        )
    )


    # Race-safe reuse.
    if cache_path.exists():

        existing_payload = load_json(
            cache_path
        )


        validate_full_relation_cache(
            batch_candidates,
            existing_payload,
        )


        return {
            "batch_id": batch_id,
            "cache_payload": (
                existing_payload
            ),
            "api_calls_this_run": 0,
            "retry_events_this_run": 0,
            "cache_created_this_run": False,
        }


    last_error = None


    for attempt in range(
        1,
        FULL_RELATION_MAX_ATTEMPTS + 1,
    ):

        (
            system_message,
            user_message,
        ) = build_relation_pilot_prompt(
            batch_candidates,
            retry_feedback=(
                last_error
            ),
        )


        try:

            wait_for_full_relation_slot()


            content = (
                call_semantic_v3_judge(
                    system_message,
                    user_message,
                )
            )


            parsed_payload = (
                parse_model_json(
                    content
                )
            )


            (
                validated_judgments,
                normalization_counts,
            ) = (
                validate_full_relation_payload(
                    batch_candidates,
                    parsed_payload,
                )
            )


            cache_payload = {
                "cache_version": (
                    FULL_RELATION_CACHE_VERSION
                ),

                "full_relation_version": (
                    FULL_RELATION_VERSION
                ),

                "batch_id": (
                    batch_id
                ),

                "semantic_model": (
                    pilot_chat_model
                ),

                "relation_prompt_version": (
                    RELATION_PROMPT_VERSION
                ),

                "source_selected_candidates_sha256": (
                    semantic_selected_candidates_sha256
                ),

                "candidate_pair_ids": [
                    candidate[
                        "candidate_pair_id"
                    ]
                    for candidate in (
                        batch_candidates
                    )
                ],

                "attempt_count": (
                    attempt
                ),

                "retry_events": (
                    attempt - 1
                ),

                "schema_normalizations": (
                    normalization_counts
                ),

                "judgments": (
                    validated_judgments
                ),
            }


            cache_text = (
                json.dumps(
                    cache_payload,
                    ensure_ascii=False,
                    indent=2,
                    sort_keys=True,
                )
                + "\n"
            )


            # If a file somehow appeared between the
            # existence check and this write, verify it
            # rather than overwriting it.
            if cache_path.exists():

                disk_payload = load_json(
                    cache_path
                )


                validate_full_relation_cache(
                    batch_candidates,
                    disk_payload,
                )


                return {
                    "batch_id": batch_id,
                    "cache_payload": (
                        disk_payload
                    ),
                    "api_calls_this_run": (
                        attempt
                    ),
                    "retry_events_this_run": (
                        attempt - 1
                    ),
                    "cache_created_this_run": False,
                }


            with cache_path.open(
                "x",
                encoding="utf-8",
            ) as file:

                file.write(
                    cache_text
                )


            disk_payload = load_json(
                cache_path
            )


            validate_full_relation_cache(
                batch_candidates,
                disk_payload,
            )


            assert (
                disk_payload
                == cache_payload
            )


            return {
                "batch_id": batch_id,
                "cache_payload": (
                    disk_payload
                ),
                "api_calls_this_run": (
                    attempt
                ),
                "retry_events_this_run": (
                    attempt - 1
                ),
                "cache_created_this_run": True,
            }


        except Exception as exc:

            last_error = (
                f"{type(exc).__name__}: "
                f"{exc}"
            )


            with full_relation_print_lock:

                print(
                    "  ",
                    batch_id,
                    "| attempt",
                    attempt,
                    "INVALID |",
                    last_error,
                )


            if (
                attempt
                >= FULL_RELATION_MAX_ATTEMPTS
            ):

                raise RuntimeError(
                    "Full Relation batch failed "
                    "after "
                    f"{FULL_RELATION_MAX_ATTEMPTS} "
                    "attempts: "
                    f"{batch_id} | "
                    f"{last_error}"
                ) from exc


# =========================================================
# Parallel execution of missing batches only
# =========================================================

full_relation_api_calls_this_run = 0

full_relation_retry_events_this_run = 0

full_relation_new_caches_this_run = 0

full_relation_completed_this_run = 0


if pending_full_relation_batches:

    with ThreadPoolExecutor(
        max_workers=(
            FULL_RELATION_WORKERS
        )
    ) as executor:

        future_map = {
            executor.submit(
                adjudicate_full_relation_batch,
                batch_candidates,
            ): (
                batch_candidates
            )
            for batch_candidates in (
                pending_full_relation_batches
            )
        }


        for future in as_completed(
            future_map
        ):

            result = future.result()


            full_relation_completed_this_run += 1


            full_relation_api_calls_this_run += (
                result[
                    "api_calls_this_run"
                ]
            )


            full_relation_retry_events_this_run += (
                result[
                    "retry_events_this_run"
                ]
            )


            if (
                result[
                    "cache_created_this_run"
                ]
            ):

                full_relation_new_caches_this_run += 1


            if (
                full_relation_completed_this_run
                % 25
                == 0
                or full_relation_completed_this_run
                == len(
                    pending_full_relation_batches
                )
            ):

                with full_relation_print_lock:

                    print(
                        "[",
                        full_relation_completed_this_run,
                        "/",
                        len(
                            pending_full_relation_batches
                        ),
                        "] VALID"
                    )


# =========================================================
# Load and validate ALL Full caches after execution
# =========================================================

full_relation_cache_payloads = []


for batch_candidates in (
    full_relation_batches
):

    cache_path = (
        full_relation_cache_path(
            batch_candidates
        )
    )


    assert (
        cache_path.exists()
    ), (
        "Full Relation cache missing after "
        f"execution: {cache_path.name}"
    )


    cache_payload = load_json(
        cache_path
    )


    validate_full_relation_cache(
        batch_candidates,
        cache_payload,
    )


    full_relation_cache_payloads.append(
        cache_payload
    )


assert (
    len(
        full_relation_cache_payloads
    )
    == len(
        full_relation_batches
    )
)


# =========================================================
# Deterministic cumulative cache diagnostics
# =========================================================

recorded_api_attempts_in_full_caches = sum(
    payload[
        "attempt_count"
    ]
    for payload in (
        full_relation_cache_payloads
    )
)


recorded_retry_events_in_full_caches = sum(
    payload[
        "retry_events"
    ]
    for payload in (
        full_relation_cache_payloads
    )
)


recorded_schema_normalizations = Counter()


for payload in (
    full_relation_cache_payloads
):

    for (
        key,
        value,
    ) in (
        payload[
            "schema_normalizations"
        ].items()
    ):

        recorded_schema_normalizations[
            key
        ] += value


# =========================================================
# Build API-judgment index from caches
# =========================================================

full_api_judgment_by_id = {}


for payload in (
    full_relation_cache_payloads
):

    for judgment in (
        payload[
            "judgments"
        ]
    ):

        candidate_pair_id = (
            judgment[
                "candidate_pair_id"
            ]
        )


        assert (
            candidate_pair_id
            not in full_api_judgment_by_id
        )


        full_api_judgment_by_id[
            candidate_pair_id
        ] = judgment


assert (
    len(
        full_api_judgment_by_id
    )
    == len(
        full_api_candidates
    )
)


assert (
    set(
        full_api_judgment_by_id
    )
    == {
        candidate[
            "candidate_pair_id"
        ]
        for candidate in (
            full_api_candidates
        )
    }
)


# =========================================================
# Combine Pilot + Full judgments
# =========================================================

full_relation_records = []


for candidate in (
    full_relation_candidates
):

    candidate_pair_id = (
        candidate[
            "candidate_pair_id"
        ]
    )


    if (
        candidate_pair_id
        in pilot_relation_by_id
    ):

        pilot_record = (
            pilot_relation_by_id[
                candidate_pair_id
            ]
        )


        judgment = {
            "decision": (
                pilot_record[
                    "decision"
                ]
            ),

            "subsumption_direction": (
                pilot_record[
                    "subsumption_direction"
                ]
            ),

            "brief_reason_fa": (
                pilot_record[
                    "brief_reason_fa"
                ]
            ),
        }


        judgment_origin = (
            "PILOT_REUSED"
        )


    else:

        api_judgment = (
            full_api_judgment_by_id[
                candidate_pair_id
            ]
        )


        judgment = {
            "decision": (
                api_judgment[
                    "decision"
                ]
            ),

            "subsumption_direction": (
                api_judgment[
                    "subsumption_direction"
                ]
            ),

            "brief_reason_fa": (
                api_judgment[
                    "brief_reason_fa"
                ]
            ),
        }


        judgment_origin = (
            "FULL_ADJUDICATION_CACHE"
        )


    full_relation_records.append({
        "candidate_pair_id": (
            candidate_pair_id
        ),

        "text_id_a": (
            candidate[
                "text_id_a"
            ]
        ),

        "text_id_b": (
            candidate[
                "text_id_b"
            ]
        ),

        "raw_claim_ids_a": (
            candidate[
                "raw_claim_ids_a"
            ]
        ),

        "raw_claim_ids_b": (
            candidate[
                "raw_claim_ids_b"
            ]
        ),

        "claim_text_a": (
            candidate[
                "claim_text_a"
            ]
        ),

        "claim_text_b": (
            candidate[
                "claim_text_b"
            ]
        ),

        "claim_types_a": (
            candidate[
                "claim_types_a"
            ]
        ),

        "claim_types_b": (
            candidate[
                "claim_types_b"
            ]
        ),

        "source_types_a": (
            candidate[
                "source_types_a"
            ]
        ),

        "source_types_b": (
            candidate[
                "source_types_b"
            ]
        ),

        "authority_classes_a": (
            candidate[
                "authority_classes_a"
            ]
        ),

        "authority_classes_b": (
            candidate[
                "authority_classes_b"
            ]
        ),

        "cosine_similarity": (
            candidate[
                "cosine_similarity"
            ]
        ),

        "mutual_top_k": (
            candidate[
                "mutual_top_k"
            ]
        ),

        "similarity_band": (
            candidate[
                "similarity_band"
            ]
        ),

        "decision": (
            judgment[
                "decision"
            ]
        ),

        "subsumption_direction": (
            judgment[
                "subsumption_direction"
            ]
        ),

        "brief_reason_fa": (
            judgment[
                "brief_reason_fa"
            ]
        ),

        "semantic_model": (
            pilot_chat_model
        ),

        "relation_prompt_version": (
            RELATION_PROMPT_VERSION
        ),

        "judgment_origin": (
            judgment_origin
        ),

        "full_relation_version": (
            FULL_RELATION_VERSION
        ),
    })


full_relation_records = sorted(
    full_relation_records,
    key=lambda item: (
        item[
            "candidate_pair_id"
        ]
    ),
)


# =========================================================
# Final structural validation
# =========================================================

assert (
    len(
        full_relation_records
    )
    == len(
        full_relation_candidates
    )
)


assert (
    len({
        item[
            "candidate_pair_id"
        ]
        for item in (
            full_relation_records
        )
    })
    == len(
        full_relation_records
    )
)


for item in (
    full_relation_records
):

    assert (
        item[
            "decision"
        ]
        in RELATION_DECISIONS
    )


    if (
        item[
            "decision"
        ]
        == "SUBSUMES"
    ):

        assert (
            item[
                "subsumption_direction"
            ]
            in SUBSUMPTION_DIRECTIONS
        )


    else:

        assert (
            item[
                "subsumption_direction"
            ]
            is None
        )


# =========================================================
# Final relation diagnostics
# =========================================================

full_relation_decision_counts = Counter(
    item[
        "decision"
    ]
    for item in (
        full_relation_records
    )
)


full_relation_by_band = defaultdict(
    Counter
)


for item in (
    full_relation_records
):

    full_relation_by_band[
        item[
            "similarity_band"
        ]
    ][
        item[
            "decision"
        ]
    ] += 1


full_relation_origin_counts = Counter(
    item[
        "judgment_origin"
    ]
    for item in (
        full_relation_records
    )
)


semantic_relation_pairs = sum(
    1
    for item in (
        full_relation_records
    )
    if (
        item[
            "decision"
        ]
        != "DISTINCT"
    )
)


cross_source_all_pairs = sum(
    1
    for item in (
        full_relation_records
    )
    if (
        set(
            item[
                "source_types_a"
            ]
        )
        != set(
            item[
                "source_types_b"
            ]
        )
    )
)


cross_source_semantic_relation_pairs = sum(
    1
    for item in (
        full_relation_records
    )
    if (
        item[
            "decision"
        ]
        != "DISTINCT"
        and (
            set(
                item[
                    "source_types_a"
                ]
            )
            != set(
                item[
                    "source_types_b"
                ]
            )
        )
    )
)


# =========================================================
# Serialize Full Adjudication artifact
# =========================================================

full_relation_artifact_text = "".join(
    json.dumps(
        item,
        ensure_ascii=False,
        sort_keys=True,
    )
    + "\n"
    for item in (
        full_relation_records
    )
)


full_relation_artifact_sha256 = (
    hashlib.sha256(
        full_relation_artifact_text.encode(
            "utf-8"
        )
    ).hexdigest()
)


full_relation_artifact_write_status = (
    publish_frozen_artifact(
        full_relation_artifact_path,
        full_relation_artifact_text,
    )
)


# =========================================================
# Deterministic frozen report
#
# IMPORTANT:
# Current-run counters are NOT stored in this report.
# Only values reconstructable from frozen artifacts/caches
# are included, so reruns remain byte-identical.
# =========================================================

full_relation_report = {
    "report_version": (
        "claim_semantic_relation_full_report_v1"
    ),

    "full_relation_version": (
        FULL_RELATION_VERSION
    ),

    "relation_prompt_version": (
        RELATION_PROMPT_VERSION
    ),

    "semantic_model": (
        pilot_chat_model
    ),

    "source_selected_candidates": (
        semantic_selected_candidates_path.name
    ),

    "source_selected_candidates_sha256": (
        semantic_selected_candidates_sha256
    ),

    "source_pilot": (
        relation_pilot_path.name
    ),

    "source_pilot_sha256": (
        relation_pilot_sha256
    ),

    "selected_pairs": (
        len(
            full_relation_records
        )
    ),

    "pilot_pairs_reused": (
        len(
            full_relation_pilot_records
        )
    ),

    "full_api_adjudicated_pairs": (
        len(
            full_api_candidates
        )
    ),

    "full_relation_batches": (
        len(
            full_relation_batches
        )
    ),

    "completed_cache_files": (
        len(
            full_relation_cache_payloads
        )
    ),

    "recorded_api_attempts_in_completed_caches": (
        recorded_api_attempts_in_full_caches
    ),

    "recorded_retry_events_in_completed_caches": (
        recorded_retry_events_in_full_caches
    ),

    "recorded_schema_normalizations": dict(
        sorted(
            recorded_schema_normalizations.items()
        )
    ),

    "decision_counts": dict(
        sorted(
            full_relation_decision_counts.items()
        )
    ),

    "decisions_by_similarity_band": {
        band: dict(
            sorted(
                counts.items()
            )
        )
        for (
            band,
            counts,
        ) in sorted(
            full_relation_by_band.items()
        )
    },

    "judgment_origin_counts": dict(
        sorted(
            full_relation_origin_counts.items()
        )
    ),

    "semantic_relation_pairs": (
        semantic_relation_pairs
    ),

    "cross_source_selected_pairs": (
        cross_source_all_pairs
    ),

    "cross_source_semantic_relation_pairs": (
        cross_source_semantic_relation_pairs
    ),

    "automatic_semantic_merge_applied": False,

    "automatic_subsumption_merge_applied": False,

    "claim_ids_assigned": False,

    "next_stage": (
        "DEDUP_RELATION_CONSOLIDATION"
    ),

    "artifact": {
        "full_adjudication": (
            full_relation_artifact_path.name
        ),

        "full_adjudication_sha256": (
            full_relation_artifact_sha256
        ),
    },
}


full_relation_report_text = (
    json.dumps(
        full_relation_report,
        ensure_ascii=False,
        indent=2,
        sort_keys=True,
    )
    + "\n"
)


full_relation_report_write_status = (
    publish_frozen_artifact(
        full_relation_report_path,
        full_relation_report_text,
    )
)


# =========================================================
# Independent disk revalidation
# =========================================================

assert (
    load_jsonl(
        full_relation_artifact_path
    )
    == full_relation_records
)


assert (
    load_json(
        full_relation_report_path
    )
    == full_relation_report
)


assert (
    file_sha256(
        full_relation_artifact_path
    )
    == full_relation_artifact_sha256
)


assert (
    full_relation_report[
        "artifact"
    ][
        "full_adjudication_sha256"
    ]
    == full_relation_artifact_sha256
)


assert (
    full_relation_report[
        "selected_pairs"
    ]
    == len(
        full_relation_candidates
    )
)


assert (
    full_relation_report[
        "pilot_pairs_reused"
    ]
    + full_relation_report[
        "full_api_adjudicated_pairs"
    ]
    == full_relation_report[
        "selected_pairs"
    ]
)


# =========================================================
# Gate
# =========================================================

semantic_relation_full_status = (
    "PASS"
)


dedup_relation_consolidation_readiness = (
    "READY_FOR_DEDUP_RELATION_CONSOLIDATION"
)


# =========================================================
# Result
# =========================================================

print()

print(
    "=== NB12i FULL SEMANTIC RELATION RESULT ==="
)

print()

print(
    "Selected pairs adjudicated:",
    len(
        full_relation_records
    ),
)

print(
    "Pilot judgments reused:",
    len(
        full_relation_pilot_records
    ),
)

print(
    "Full API-adjudicated pairs:",
    len(
        full_api_candidates
    ),
)


print()

print(
    "Full relation caches:",
    len(
        full_relation_cache_payloads
    ),
)

print(
    "API calls THIS run:",
    full_relation_api_calls_this_run,
)

print(
    "Retry events THIS run:",
    full_relation_retry_events_this_run,
)

print(
    "New caches THIS run:",
    full_relation_new_caches_this_run,
)


print()

print(
    "Recorded API attempts in completed caches:",
    recorded_api_attempts_in_full_caches,
)

print(
    "Recorded retry events in completed caches:",
    recorded_retry_events_in_full_caches,
)

print(
    "Recorded schema normalizations:",
    dict(
        sorted(
            recorded_schema_normalizations.items()
        )
    ),
)


print()

print(
    "Relation decisions:",
    dict(
        sorted(
            full_relation_decision_counts.items()
        )
    ),
)

print(
    "Semantic relation pairs "
    "(non-DISTINCT):",
    semantic_relation_pairs,
)

print(
    "Cross-source selected pairs:",
    cross_source_all_pairs,
)

print(
    "Cross-source semantic relation pairs:",
    cross_source_semantic_relation_pairs,
)


print()

print(
    "Decisions by similarity band:"
)


for band in (
    expected_similarity_bands
):

    print(
        " ",
        band,
        ":",
        dict(
            sorted(
                full_relation_by_band[
                    band
                ].items()
            )
        ),
    )


print()

print(
    "Automatic semantic merge applied:",
    False,
)

print(
    "Automatic subsumption merge applied:",
    False,
)

print(
    "Claim IDs assigned:",
    False,
)


print()

print(
    "Full Adjudication Artifact:",
    full_relation_artifact_path.relative_to(
        project_root
    ),
)

print(
    "Full Adjudication SHA-256:",
    full_relation_artifact_sha256,
)

print(
    "Artifact write status:",
    full_relation_artifact_write_status,
)

print(
    "Report write status:",
    full_relation_report_write_status,
)


print()

print(
    "Next stage:",
    "DEDUP_RELATION_CONSOLIDATION",
)

print(
    "Dedup consolidation readiness:",
    dedup_relation_consolidation_readiness,
)


print()

print(
    "CODE CELL 25 GATE:",
    semantic_relation_full_status,
)
```

<div style="direction: rtl; white-space: normal; line-height: 1.7;">

<h2>Consolidation روابط Dedup</h2>

<p>
داوری کامل Semantic Relation با موفقیت پایان یافته است.
در این مرحله روابط Exact و Semantic در یک Ledger واحد و deterministic
تجمیع می‌شوند.
</p>

<p>
تصمیم <code>DISTINCT</code> یک Relation محسوب نمی‌شود و در Ledger نهایی
روابط وارد نمی‌شود. فقط روابط
<code>EXACT_DUPLICATE</code>،
<code>SEMANTIC_DUPLICATE</code>،
<code>SUBSUMES</code> و
<code>PARTIAL_OVERLAP</code>
ثبت می‌شوند.
</p>

<ul>
<li>هیچ API یا Qwen استفاده نمی‌شود.</li>
<li>هیچ Semantic Duplicate به‌طور خودکار Merge نمی‌شود.</li>
<li>هیچ Subsumption به‌طور خودکار Merge نمی‌شود.</li>
<li>هویت Raw Claimها و Evidence multiplicity حفظ می‌شود.</li>
<li>Claim ID نهایی هنوز در این مرحله ساخته نمی‌شود.</li>
</ul>

</div>

```python
# Code Cell 26

assert (
    semantic_relation_full_status
    == "PASS"
), (
    "Code Cell 25 must PASS before "
    "Dedup Relation Consolidation."
)

assert (
    dedup_relation_consolidation_readiness
    == "READY_FOR_DEDUP_RELATION_CONSOLIDATION"
)


# =========================================================
# Configuration
# =========================================================

DEDUP_RELATION_LEDGER_VERSION = (
    "claim_dedup_relation_ledger_v1"
)

ALLOWED_DEDUP_RELATION_TYPES = {
    "EXACT_DUPLICATE",
    "SEMANTIC_DUPLICATE",
    "SUBSUMES",
    "PARTIAL_OVERLAP",
}


dedup_relation_ledger_path = (
    registry_artifact_root
    / "claim_dedup_relations_consolidated_v1.jsonl"
)

dedup_relation_consolidation_report_path = (
    registry_artifact_root
    / "claim_dedup_relation_consolidation_report_v1.json"
)


# =========================================================
# Independently reload frozen inputs
# =========================================================

consolidation_validated_claims = load_jsonl(
    validated_claims_path
)

consolidation_full_relations = load_jsonl(
    full_relation_artifact_path
)


validated_claims_disk_sha256 = (
    file_sha256(
        validated_claims_path
    )
)

full_relation_disk_sha256 = (
    file_sha256(
        full_relation_artifact_path
    )
)


assert (
    full_relation_disk_sha256
    == full_relation_artifact_sha256
), (
    "Full Semantic Relation artifact changed."
)


assert (
    len(
        consolidation_full_relations
    )
    == 3173
)


# =========================================================
# Validated Raw Claim index
# =========================================================

validated_claim_by_id = {
    claim[
        "raw_claim_id"
    ]: claim
    for claim in (
        consolidation_validated_claims
    )
}


assert (
    len(
        validated_claim_by_id
    )
    == len(
        consolidation_validated_claims
    )
), (
    "Duplicate raw_claim_id in validated Claims."
)


# =========================================================
# Rebuild Exact Duplicate groups independently from
# validated Claim text.
#
# This avoids depending on the internal row schema of
# Cell 20 while preserving the exact same deterministic
# definition: stored claim_text_fa equality.
# =========================================================

raw_claim_ids_by_exact_text = defaultdict(
    list
)


for claim in (
    consolidation_validated_claims
):

    claim_text = claim[
        "claim_text_fa"
    ]

    raw_claim_ids_by_exact_text[
        claim_text
    ].append(
        claim[
            "raw_claim_id"
        ]
    )


for claim_text in (
    raw_claim_ids_by_exact_text
):

    raw_claim_ids_by_exact_text[
        claim_text
    ] = sorted(
        raw_claim_ids_by_exact_text[
            claim_text
        ]
    )


exact_duplicate_text_groups = {
    claim_text: raw_claim_ids
    for (
        claim_text,
        raw_claim_ids,
    ) in (
        raw_claim_ids_by_exact_text.items()
    )
    if (
        len(
            raw_claim_ids
        )
        > 1
    )
}


exact_unique_text_count = len(
    raw_claim_ids_by_exact_text
)

exact_duplicate_group_count = len(
    exact_duplicate_text_groups
)

claims_inside_exact_groups = sum(
    len(
        raw_claim_ids
    )
    for raw_claim_ids in (
        exact_duplicate_text_groups.values()
    )
)


# =========================================================
# Helper: stable text ID
# =========================================================

def consolidated_text_id(
    claim_text,
):

    digest = hashlib.sha256(
        claim_text.encode(
            "utf-8"
        )
    ).hexdigest()

    return (
        "claim-text:"
        + digest[:24]
    )


# =========================================================
# Build EXACT_DUPLICATE relation rows.
#
# These remain raw-claim pair relations because the
# eventual exact-canonical Claim node has not yet been
# assigned.
# =========================================================

exact_relation_rows = []


for (
    claim_text,
    raw_claim_ids,
) in sorted(
    exact_duplicate_text_groups.items(),
    key=lambda item: (
        consolidated_text_id(
            item[0]
        )
    ),
):

    text_id = consolidated_text_id(
        claim_text
    )


    for left_index in range(
        len(
            raw_claim_ids
        )
    ):

        for right_index in range(
            left_index + 1,
            len(
                raw_claim_ids
            ),
        ):

            raw_claim_id_a = (
                raw_claim_ids[
                    left_index
                ]
            )

            raw_claim_id_b = (
                raw_claim_ids[
                    right_index
                ]
            )


            relation_material = (
                "EXACT_DUPLICATE"
                + "\n"
                + raw_claim_id_a
                + "\n"
                + raw_claim_id_b
            )


            relation_digest = hashlib.sha256(
                relation_material.encode(
                    "utf-8"
                )
            ).hexdigest()


            exact_relation_rows.append({
                "relation_id": (
                    "claim-relation:"
                    + relation_digest[:24]
                ),

                "relation_type": (
                    "EXACT_DUPLICATE"
                ),

                "relation_scope": (
                    "RAW_CLAIM_PAIR"
                ),

                "candidate_pair_id": None,

                "text_id_a": (
                    text_id
                ),

                "text_id_b": (
                    text_id
                ),

                "raw_claim_ids_a": [
                    raw_claim_id_a
                ],

                "raw_claim_ids_b": [
                    raw_claim_id_b
                ],

                "claim_text_a": (
                    claim_text
                ),

                "claim_text_b": (
                    claim_text
                ),

                "cosine_similarity": None,

                "subsumption_direction": None,

                "judgment_origin": (
                    "DETERMINISTIC_EXACT_TEXT"
                ),

                "source_stage": (
                    "EXACT_DEDUP_STAGE_1"
                ),

                "ledger_version": (
                    DEDUP_RELATION_LEDGER_VERSION
                ),
            })


# =========================================================
# Validate Exact relations
# =========================================================

expected_exact_pair_count = sum(
    (
        len(
            raw_claim_ids
        )
        * (
            len(
                raw_claim_ids
            )
            - 1
        )
    )
    // 2
    for raw_claim_ids in (
        exact_duplicate_text_groups.values()
    )
)


assert (
    len(
        exact_relation_rows
    )
    == expected_exact_pair_count
)


for relation in (
    exact_relation_rows
):

    assert (
        relation[
            "relation_type"
        ]
        == "EXACT_DUPLICATE"
    )

    assert (
        relation[
            "claim_text_a"
        ]
        == relation[
            "claim_text_b"
        ]
    )

    assert (
        relation[
            "raw_claim_ids_a"
        ][0]
        != relation[
            "raw_claim_ids_b"
        ][0]
    )


# =========================================================
# Build Semantic Relation rows.
#
# DISTINCT is intentionally excluded because it is not a
# relation in the frozen Registry contract.
# =========================================================

semantic_relation_rows = []

semantic_distinct_count = 0


for item in (
    consolidation_full_relations
):

    decision = item[
        "decision"
    ]


    if (
        decision
        == "DISTINCT"
    ):

        semantic_distinct_count += 1

        continue


    assert (
        decision
        in {
            "SEMANTIC_DUPLICATE",
            "SUBSUMES",
            "PARTIAL_OVERLAP",
        }
    )


    if (
        decision
        == "SUBSUMES"
    ):

        assert (
            item[
                "subsumption_direction"
            ]
            in SUBSUMPTION_DIRECTIONS
        )

    else:

        assert (
            item[
                "subsumption_direction"
            ]
            is None
        )


    relation_material = (
        decision
        + "\n"
        + item[
            "candidate_pair_id"
        ]
        + "\n"
        + str(
            item[
                "subsumption_direction"
            ]
        )
    )


    relation_digest = hashlib.sha256(
        relation_material.encode(
            "utf-8"
        )
    ).hexdigest()


    semantic_relation_rows.append({
        "relation_id": (
            "claim-relation:"
            + relation_digest[:24]
        ),

        "relation_type": (
            decision
        ),

        "relation_scope": (
            "EXACT_UNIQUE_TEXT_PAIR"
        ),

        "candidate_pair_id": (
            item[
                "candidate_pair_id"
            ]
        ),

        "text_id_a": (
            item[
                "text_id_a"
            ]
        ),

        "text_id_b": (
            item[
                "text_id_b"
            ]
        ),

        "raw_claim_ids_a": (
            item[
                "raw_claim_ids_a"
            ]
        ),

        "raw_claim_ids_b": (
            item[
                "raw_claim_ids_b"
            ]
        ),

        "claim_text_a": (
            item[
                "claim_text_a"
            ]
        ),

        "claim_text_b": (
            item[
                "claim_text_b"
            ]
        ),

        "cosine_similarity": (
            item[
                "cosine_similarity"
            ]
        ),

        "subsumption_direction": (
            item[
                "subsumption_direction"
            ]
        ),

        "judgment_origin": (
            item[
                "judgment_origin"
            ]
        ),

        "source_stage": (
            "FULL_SEMANTIC_RELATION_ADJUDICATION"
        ),

        "ledger_version": (
            DEDUP_RELATION_LEDGER_VERSION
        ),
    })


# =========================================================
# Semantic diagnostics
# =========================================================

semantic_relation_type_counts = Counter(
    relation[
        "relation_type"
    ]
    for relation in (
        semantic_relation_rows
    )
)


assert (
    sum(
        semantic_relation_type_counts.values()
    )
    + semantic_distinct_count
    == len(
        consolidation_full_relations
    )
)


assert (
    semantic_relation_type_counts[
        "SEMANTIC_DUPLICATE"
    ]
    == full_relation_decision_counts[
        "SEMANTIC_DUPLICATE"
    ]
)


assert (
    semantic_relation_type_counts[
        "SUBSUMES"
    ]
    == full_relation_decision_counts[
        "SUBSUMES"
    ]
)


assert (
    semantic_relation_type_counts[
        "PARTIAL_OVERLAP"
    ]
    == full_relation_decision_counts[
        "PARTIAL_OVERLAP"
    ]
)


assert (
    semantic_distinct_count
    == full_relation_decision_counts[
        "DISTINCT"
    ]
)


# =========================================================
# Consolidated Relation Ledger
# =========================================================

dedup_relation_ledger = (
    exact_relation_rows
    + semantic_relation_rows
)


dedup_relation_ledger = sorted(
    dedup_relation_ledger,
    key=lambda item: (
        item[
            "relation_type"
        ],
        item[
            "relation_id"
        ],
    ),
)


relation_ids = [
    item[
        "relation_id"
    ]
    for item in (
        dedup_relation_ledger
    )
]


assert (
    len(
        relation_ids
    )
    == len(
        set(
            relation_ids
        )
    )
), (
    "Duplicate relation_id detected."
)


assert all(
    relation[
        "relation_type"
    ]
    in ALLOWED_DEDUP_RELATION_TYPES
    for relation in (
        dedup_relation_ledger
    )
)


assert all(
    raw_claim_id
    in validated_claim_by_id
    for relation in (
        dedup_relation_ledger
    )
    for raw_claim_id in (
        relation[
            "raw_claim_ids_a"
        ]
        + relation[
            "raw_claim_ids_b"
        ]
    )
)


# =========================================================
# Final relation counts
# =========================================================

consolidated_relation_type_counts = Counter(
    item[
        "relation_type"
    ]
    for item in (
        dedup_relation_ledger
    )
)


semantic_duplicate_relation_count = (
    consolidated_relation_type_counts[
        "SEMANTIC_DUPLICATE"
    ]
)

subsumes_relation_count = (
    consolidated_relation_type_counts[
        "SUBSUMES"
    ]
)

partial_overlap_relation_count = (
    consolidated_relation_type_counts[
        "PARTIAL_OVERLAP"
    ]
)

exact_duplicate_relation_count = (
    consolidated_relation_type_counts[
        "EXACT_DUPLICATE"
    ]
)


# =========================================================
# Serialize frozen Relation Ledger
# =========================================================

dedup_relation_ledger_text = "".join(
    json.dumps(
        item,
        ensure_ascii=False,
        sort_keys=True,
    )
    + "\n"
    for item in (
        dedup_relation_ledger
    )
)


dedup_relation_ledger_sha256 = (
    hashlib.sha256(
        dedup_relation_ledger_text.encode(
            "utf-8"
        )
    ).hexdigest()
)


dedup_relation_ledger_write_status = (
    publish_frozen_artifact(
        dedup_relation_ledger_path,
        dedup_relation_ledger_text,
    )
)


# =========================================================
# Deterministic consolidation report
# =========================================================

dedup_relation_consolidation_report = {
    "report_version": (
        "claim_dedup_relation_consolidation_report_v1"
    ),

    "ledger_version": (
        DEDUP_RELATION_LEDGER_VERSION
    ),

    "source_validated_claims": (
        validated_claims_path.name
    ),

    "source_validated_claims_sha256": (
        validated_claims_disk_sha256
    ),

    "source_exact_dedup_report": (
        dedup_stage1_report_path.name
    ),

    "source_exact_dedup_report_sha256": (
        file_sha256(
            dedup_stage1_report_path
        )
    ),

    "source_full_semantic_relations": (
        full_relation_artifact_path.name
    ),

    "source_full_semantic_relations_sha256": (
        full_relation_disk_sha256
    ),

    "validated_raw_claims": (
        len(
            consolidation_validated_claims
        )
    ),

    "exact_unique_claim_texts": (
        exact_unique_text_count
    ),

    "exact_duplicate_groups": (
        exact_duplicate_group_count
    ),

    "claims_inside_exact_duplicate_groups": (
        claims_inside_exact_groups
    ),

    "exact_duplicate_pair_relations": (
        exact_duplicate_relation_count
    ),

    "semantic_pairs_adjudicated": (
        len(
            consolidation_full_relations
        )
    ),

    "semantic_distinct_pairs_excluded": (
        semantic_distinct_count
    ),

    "semantic_relation_counts": dict(
        sorted(
            semantic_relation_type_counts.items()
        )
    ),

    "consolidated_relation_counts": dict(
        sorted(
            consolidated_relation_type_counts.items()
        )
    ),

    "consolidated_relation_records": (
        len(
            dedup_relation_ledger
        )
    ),

    "automatic_semantic_merge_applied": False,

    "automatic_subsumption_merge_applied": False,

    "exact_claim_canonicalization_applied": False,

    "claim_ids_assigned": False,

    "api_calls": 0,

    "next_stage": (
        "EXACT_CANONICAL_REGISTRY_AND_CLAIM_ID_ASSIGNMENT"
    ),

    "artifact": {
        "relation_ledger": (
            dedup_relation_ledger_path.name
        ),

        "relation_ledger_sha256": (
            dedup_relation_ledger_sha256
        ),
    },
}


dedup_relation_consolidation_report_text = (
    json.dumps(
        dedup_relation_consolidation_report,
        ensure_ascii=False,
        indent=2,
        sort_keys=True,
    )
    + "\n"
)


dedup_relation_consolidation_report_write_status = (
    publish_frozen_artifact(
        dedup_relation_consolidation_report_path,
        dedup_relation_consolidation_report_text,
    )
)


# =========================================================
# Independent disk revalidation
# =========================================================

assert (
    load_jsonl(
        dedup_relation_ledger_path
    )
    == dedup_relation_ledger
)


assert (
    load_json(
        dedup_relation_consolidation_report_path
    )
    == dedup_relation_consolidation_report
)


assert (
    file_sha256(
        dedup_relation_ledger_path
    )
    == dedup_relation_ledger_sha256
)


assert (
    dedup_relation_consolidation_report[
        "artifact"
    ][
        "relation_ledger_sha256"
    ]
    == dedup_relation_ledger_sha256
)


# =========================================================
# Gate
# =========================================================

dedup_relation_consolidation_status = (
    "PASS"
)


exact_canonical_registry_readiness = (
    "READY_FOR_EXACT_CANONICAL_REGISTRY_AND_CLAIM_ID_ASSIGNMENT"
)


# =========================================================
# Result
# =========================================================

print(
    "=== NB12i DEDUP RELATION CONSOLIDATION ==="
)

print()

print(
    "Validated Raw Claims:",
    len(
        consolidation_validated_claims
    ),
)

print(
    "Exact-unique Claim texts:",
    exact_unique_text_count,
)

print(
    "Exact Duplicate groups:",
    exact_duplicate_group_count,
)

print(
    "Claims inside Exact groups:",
    claims_inside_exact_groups,
)


print()

print(
    "Exact Duplicate pair relations:",
    exact_duplicate_relation_count,
)

print(
    "Semantic pairs adjudicated:",
    len(
        consolidation_full_relations
    ),
)

print(
    "DISTINCT pairs excluded:",
    semantic_distinct_count,
)


print()

print(
    "Semantic relation counts:",
    dict(
        sorted(
            semantic_relation_type_counts.items()
        )
    ),
)

print(
    "Consolidated relation counts:",
    dict(
        sorted(
            consolidated_relation_type_counts.items()
        )
    ),
)

print(
    "Consolidated Relation records:",
    len(
        dedup_relation_ledger
    ),
)


print()

print(
    "Automatic semantic merge applied:",
    False,
)

print(
    "Automatic subsumption merge applied:",
    False,
)

print(
    "Exact canonicalization applied:",
    False,
)

print(
    "Claim IDs assigned:",
    False,
)

print(
    "API calls performed:",
    0,
)


print()

print(
    "Relation Ledger:",
    dedup_relation_ledger_path.relative_to(
        project_root
    ),
)

print(
    "Relation Ledger SHA-256:",
    dedup_relation_ledger_sha256,
)

print(
    "Ledger write status:",
    dedup_relation_ledger_write_status,
)

print(
    "Report write status:",
    dedup_relation_consolidation_report_write_status,
)


print()

print(
    "Next stage:",
    "EXACT_CANONICAL_REGISTRY_AND_CLAIM_ID_ASSIGNMENT",
)

print(
    "Registry readiness:",
    exact_canonical_registry_readiness,
)


print()

print(
    "CODE CELL 26 GATE:",
    dedup_relation_consolidation_status,
)
```

<div style="direction: rtl; white-space: normal; line-height: 1.7;">

<h2>ساخت Canonical Claim Registry و تخصیص Claim ID</h2>

<p>
پس از تکمیل Validation و Dedup Relationها، اکنون هویت نهایی Claimهای Registry ساخته می‌شود.
فقط Exact Duplicateها که متن Claim کاملاً یکسان دارند در یک Canonical Claim قرار می‌گیرند.
</p>

<p>
Semantic Duplicateها، Subsumptionها و Partial Overlapها در این مرحله Merge نمی‌شوند
و به‌عنوان Claimهای مستقل با Relationهای جداگانه حفظ خواهند شد.
</p>

<ul>
<li>هر متن Exact-Unique یک <code>claim_id</code> پایدار دریافت می‌کند.</li>
<li>تمام <code>raw_claim_id</code>های سازنده حفظ می‌شوند.</li>
<li>تمام Evidenceها و Source identityها حفظ می‌شوند.</li>
<li>Metadataهای متفاوت به‌صورت مجموعه نگهداری می‌شوند و مقدار ساختگی انتخاب نمی‌شود.</li>
<li>هیچ API یا مدل زبانی استفاده نمی‌شود.</li>
<li>خروجی اصلی این مرحله <code>claims_deduplicated_v1.jsonl</code> است.</li>
</ul>

</div>

```python
# Code Cell 27

assert (
    dedup_relation_consolidation_status
    == "PASS"
), (
    "Code Cell 26 must PASS before "
    "Canonical Registry construction."
)

assert (
    exact_canonical_registry_readiness
    == "READY_FOR_EXACT_CANONICAL_REGISTRY_AND_CLAIM_ID_ASSIGNMENT"
)


# =========================================================
# Configuration
# =========================================================

CANONICAL_REGISTRY_VERSION = (
    "claim_exact_canonical_registry_v1"
)

CLAIM_ID_VERSION = (
    "claim_id_sha256_exact_text_v1"
)


deduplicated_claims_path = (
    registry_artifact_root
    / "claims_deduplicated_v1.jsonl"
)

claim_id_mapping_path = (
    registry_artifact_root
    / "claim_id_mapping_v1.jsonl"
)

canonical_registry_report_path = (
    registry_artifact_root
    / "claim_exact_canonicalization_report_v1.json"
)


# =========================================================
# Independently reload frozen inputs
# =========================================================

canonical_validated_claims = load_jsonl(
    validated_claims_path
)

canonical_evidence_links = load_jsonl(
    full_evidence_path
)

canonical_relation_ledger = load_jsonl(
    dedup_relation_ledger_path
)

canonical_consolidation_report = load_json(
    dedup_relation_consolidation_report_path
)


validated_claims_current_sha256 = (
    file_sha256(
        validated_claims_path
    )
)

evidence_links_current_sha256 = (
    file_sha256(
        full_evidence_path
    )
)

relation_ledger_current_sha256 = (
    file_sha256(
        dedup_relation_ledger_path
    )
)


assert (
    relation_ledger_current_sha256
    == dedup_relation_ledger_sha256
), (
    "Consolidated Relation Ledger changed."
)


assert (
    len(
        canonical_validated_claims
    )
    == canonical_consolidation_report[
        "validated_raw_claims"
    ]
)


# =========================================================
# Raw Claim index
# =========================================================

canonical_raw_claim_by_id = {
    claim[
        "raw_claim_id"
    ]: claim
    for claim in (
        canonical_validated_claims
    )
}


assert (
    len(
        canonical_raw_claim_by_id
    )
    == len(
        canonical_validated_claims
    )
), (
    "Duplicate raw_claim_id detected."
)


# =========================================================
# Evidence index for VALIDATED Claims only
# =========================================================

evidence_by_raw_claim_id = defaultdict(
    list
)


seen_evidence_ids = set()


for evidence in (
    canonical_evidence_links
):

    evidence_id = evidence[
        "evidence_id"
    ]


    assert (
        evidence_id
        not in seen_evidence_ids
    ), (
        "Duplicate evidence_id detected: "
        f"{evidence_id}"
    )


    seen_evidence_ids.add(
        evidence_id
    )


    raw_claim_id = evidence[
        "raw_claim_id"
    ]


    if (
        raw_claim_id
        in canonical_raw_claim_by_id
    ):

        evidence_by_raw_claim_id[
            raw_claim_id
        ].append(
            evidence
        )


for raw_claim_id in (
    canonical_raw_claim_by_id
):

    assert (
        evidence_by_raw_claim_id[
            raw_claim_id
        ]
    ), (
        "Validated Claim has no Evidence: "
        f"{raw_claim_id}"
    )


# =========================================================
# Exact-text groups
# =========================================================

raw_claim_ids_by_canonical_text = defaultdict(
    list
)


for claim in (
    canonical_validated_claims
):

    claim_text = claim[
        "claim_text_fa"
    ]


    assert (
        isinstance(
            claim_text,
            str,
        )
        and claim_text.strip()
    )


    raw_claim_ids_by_canonical_text[
        claim_text
    ].append(
        claim[
            "raw_claim_id"
        ]
    )


for claim_text in (
    raw_claim_ids_by_canonical_text
):

    raw_claim_ids_by_canonical_text[
        claim_text
    ] = sorted(
        raw_claim_ids_by_canonical_text[
            claim_text
        ]
    )


assert (
    len(
        raw_claim_ids_by_canonical_text
    )
    == canonical_consolidation_report[
        "exact_unique_claim_texts"
    ]
)


# =========================================================
# Stable Claim ID
#
# Identity is based ONLY on the exact canonical Claim text.
# No semantic merge participates in Claim identity.
# =========================================================

def build_final_claim_identity(
    claim_text,
):

    claim_text_sha256 = hashlib.sha256(
        claim_text.encode(
            "utf-8"
        )
    ).hexdigest()


    claim_id = (
        "claim:"
        + claim_text_sha256
    )


    return (
        claim_id,
        claim_text_sha256,
    )


# =========================================================
# Build final canonical Claim records
# =========================================================

deduplicated_claim_records = []

claim_id_mapping_records = []


metadata_multi_value_counts = Counter()


for (
    claim_text,
    raw_claim_ids,
) in sorted(
    raw_claim_ids_by_canonical_text.items(),
    key=lambda item: (
        build_final_claim_identity(
            item[0]
        )[0]
    ),
):

    (
        claim_id,
        claim_text_sha256,
    ) = build_final_claim_identity(
        claim_text
    )


    member_claims = [
        canonical_raw_claim_by_id[
            raw_claim_id
        ]
        for raw_claim_id in (
            raw_claim_ids
        )
    ]


    # -----------------------------------------------------
    # Preserve all Claim metadata values.
    # -----------------------------------------------------

    claim_types = sorted({
        claim[
            "claim_type"
        ]
        for claim in (
            member_claims
        )
    })


    source_document_ids = sorted({
        document_id
        for claim in (
            member_claims
        )
        for document_id in (
            claim[
                "source_document_ids"
            ]
        )
    })


    authority_classes = sorted({
        claim[
            "authority_class"
        ]
        for claim in (
            member_claims
        )
    })


    claim_discovery_roles = sorted({
        claim[
            "claim_discovery_role"
        ]
        for claim in (
            member_claims
        )
    })


    extraction_unit_ids = sorted({
        claim[
            "extraction_unit_id"
        ]
        for claim in (
            member_claims
        )
    })


    # -----------------------------------------------------
    # Preserve ALL Evidence instances.
    # -----------------------------------------------------

    member_evidence = [
        evidence
        for raw_claim_id in (
            raw_claim_ids
        )
        for evidence in (
            evidence_by_raw_claim_id[
                raw_claim_id
            ]
        )
    ]


    evidence_ids = sorted({
        evidence[
            "evidence_id"
        ]
        for evidence in (
            member_evidence
        )
    })


    assert (
        len(
            evidence_ids
        )
        == len(
            member_evidence
        )
    ), (
        "Evidence multiplicity/identity conflict "
        f"for claim_id={claim_id}"
    )


    source_ids = sorted({
        evidence[
            "source_id"
        ]
        for evidence in (
            member_evidence
        )
    })


    source_types = sorted({
        evidence[
            "source_type"
        ]
        for evidence in (
            member_evidence
        )
    })


    # -----------------------------------------------------
    # Diagnostic-only metadata multiplicity.
    # No arbitrary canonical metadata value is invented.
    # -----------------------------------------------------

    if (
        len(
            claim_types
        )
        > 1
    ):

        metadata_multi_value_counts[
            "claim_type"
        ] += 1


    if (
        len(
            authority_classes
        )
        > 1
    ):

        metadata_multi_value_counts[
            "authority_class"
        ] += 1


    if (
        len(
            claim_discovery_roles
        )
        > 1
    ):

        metadata_multi_value_counts[
            "claim_discovery_role"
        ] += 1


    # -----------------------------------------------------
    # Convenience singular fields are populated ONLY when
    # the canonical group is unanimous.
    # -----------------------------------------------------

    canonical_claim_type = (
        claim_types[0]
        if (
            len(
                claim_types
            )
            == 1
        )
        else None
    )


    canonical_authority_class = (
        authority_classes[0]
        if (
            len(
                authority_classes
            )
            == 1
        )
        else None
    )


    canonical_claim_discovery_role = (
        claim_discovery_roles[0]
        if (
            len(
                claim_discovery_roles
            )
            == 1
        )
        else None
    )


    representative_raw_claim_id = (
        raw_claim_ids[0]
    )


    canonical_record = {
        "claim_id": (
            claim_id
        ),

        "claim_id_version": (
            CLAIM_ID_VERSION
        ),

        "claim_text_fa": (
            claim_text
        ),

        "claim_text_sha256": (
            claim_text_sha256
        ),

        "raw_claim_ids": (
            raw_claim_ids
        ),

        "raw_claim_count": (
            len(
                raw_claim_ids
            )
        ),

        "representative_raw_claim_id": (
            representative_raw_claim_id
        ),

        "claim_type": (
            canonical_claim_type
        ),

        "claim_types": (
            claim_types
        ),

        "source_document_ids": (
            source_document_ids
        ),

        "source_ids": (
            source_ids
        ),

        "source_types": (
            source_types
        ),

        "authority_class": (
            canonical_authority_class
        ),

        "authority_classes": (
            authority_classes
        ),

        "claim_discovery_role": (
            canonical_claim_discovery_role
        ),

        "claim_discovery_roles": (
            claim_discovery_roles
        ),

        "extraction_unit_ids": (
            extraction_unit_ids
        ),

        "evidence_ids": (
            evidence_ids
        ),

        "evidence_count": (
            len(
                evidence_ids
            )
        ),

        "exact_duplicate_collapsed": (
            len(
                raw_claim_ids
            )
            > 1
        ),

        "semantic_merge_applied": False,

        "subsumption_merge_applied": False,

        "registry_validation_status": (
            "VALIDATED"
        ),

        "registry_stage": (
            "POST_VALIDATION_EXACT_DEDUP"
        ),

        "canonicalization_rule": (
            "EXACT_STORED_CLAIM_TEXT_EQUALITY"
        ),

        "canonical_registry_version": (
            CANONICAL_REGISTRY_VERSION
        ),
    }


    deduplicated_claim_records.append(
        canonical_record
    )


    for raw_claim_id in (
        raw_claim_ids
    ):

        claim_id_mapping_records.append({
            "raw_claim_id": (
                raw_claim_id
            ),

            "claim_id": (
                claim_id
            ),

            "claim_text_sha256": (
                claim_text_sha256
            ),

            "exact_duplicate_collapsed": (
                len(
                    raw_claim_ids
                )
                > 1
            ),

            "canonical_registry_version": (
                CANONICAL_REGISTRY_VERSION
            ),
        })


# =========================================================
# Deterministic ordering
# =========================================================

deduplicated_claim_records = sorted(
    deduplicated_claim_records,
    key=lambda item: (
        item[
            "claim_id"
        ]
    ),
)


claim_id_mapping_records = sorted(
    claim_id_mapping_records,
    key=lambda item: (
        item[
            "raw_claim_id"
        ]
    ),
)


# =========================================================
# Identity validation
# =========================================================

final_claim_ids = [
    item[
        "claim_id"
    ]
    for item in (
        deduplicated_claim_records
    )
]


assert (
    len(
        final_claim_ids
    )
    == len(
        set(
            final_claim_ids
        )
    )
), (
    "Duplicate final claim_id detected."
)


final_claim_texts = [
    item[
        "claim_text_fa"
    ]
    for item in (
        deduplicated_claim_records
    )
]


assert (
    len(
        final_claim_texts
    )
    == len(
        set(
            final_claim_texts
        )
    )
), (
    "Duplicate canonical Claim text remains."
)


mapped_raw_claim_ids = [
    item[
        "raw_claim_id"
    ]
    for item in (
        claim_id_mapping_records
    )
]


assert (
    len(
        mapped_raw_claim_ids
    )
    == len(
        set(
            mapped_raw_claim_ids
        )
    )
)


assert (
    set(
        mapped_raw_claim_ids
    )
    == set(
        canonical_raw_claim_by_id
    )
), (
    "Raw Claim -> Claim ID mapping is not complete."
)


assert (
    len(
        deduplicated_claim_records
    )
    == len(
        raw_claim_ids_by_canonical_text
    )
)


# =========================================================
# Verify every Claim ID is reproducible from exact text
# =========================================================

for record in (
    deduplicated_claim_records
):

    (
        expected_claim_id,
        expected_text_sha256,
    ) = build_final_claim_identity(
        record[
            "claim_text_fa"
        ]
    )


    assert (
        record[
            "claim_id"
        ]
        == expected_claim_id
    )


    assert (
        record[
            "claim_text_sha256"
        ]
        == expected_text_sha256
    )


# =========================================================
# Exact-collapse diagnostics
# =========================================================

exact_collapsed_canonical_claims = sum(
    1
    for record in (
        deduplicated_claim_records
    )
    if record[
        "exact_duplicate_collapsed"
    ]
)


singleton_canonical_claims = sum(
    1
    for record in (
        deduplicated_claim_records
    )
    if not record[
        "exact_duplicate_collapsed"
    ]
)


raw_claims_collapsed_away = (
    len(
        canonical_validated_claims
    )
    - len(
        deduplicated_claim_records
    )
)


assert (
    exact_collapsed_canonical_claims
    == canonical_consolidation_report[
        "exact_duplicate_groups"
    ]
)


assert (
    raw_claims_collapsed_away
    == (
        len(
            canonical_validated_claims
        )
        - canonical_consolidation_report[
            "exact_unique_claim_texts"
        ]
    )
)


# =========================================================
# Evidence preservation validation
# =========================================================

validated_input_evidence_ids = {
    evidence[
        "evidence_id"
    ]
    for raw_claim_id in (
        canonical_raw_claim_by_id
    )
    for evidence in (
        evidence_by_raw_claim_id[
            raw_claim_id
        ]
    )
}


canonical_output_evidence_ids = {
    evidence_id
    for record in (
        deduplicated_claim_records
    )
    for evidence_id in (
        record[
            "evidence_ids"
        ]
    )
}


assert (
    canonical_output_evidence_ids
    == validated_input_evidence_ids
), (
    "Evidence identity was lost or added "
    "during exact canonicalization."
)


canonical_output_evidence_occurrences = sum(
    record[
        "evidence_count"
    ]
    for record in (
        deduplicated_claim_records
    )
)


validated_input_evidence_occurrences = sum(
    len(
        evidence_by_raw_claim_id[
            raw_claim_id
        ]
    )
    for raw_claim_id in (
        canonical_raw_claim_by_id
    )
)


assert (
    canonical_output_evidence_occurrences
    == validated_input_evidence_occurrences
), (
    "Evidence multiplicity was not preserved."
)


# =========================================================
# Serialize canonical Registry
# =========================================================

deduplicated_claims_text = "".join(
    json.dumps(
        item,
        ensure_ascii=False,
        sort_keys=True,
    )
    + "\n"
    for item in (
        deduplicated_claim_records
    )
)


deduplicated_claims_sha256 = (
    hashlib.sha256(
        deduplicated_claims_text.encode(
            "utf-8"
        )
    ).hexdigest()
)


claim_id_mapping_text = "".join(
    json.dumps(
        item,
        ensure_ascii=False,
        sort_keys=True,
    )
    + "\n"
    for item in (
        claim_id_mapping_records
    )
)


claim_id_mapping_sha256 = (
    hashlib.sha256(
        claim_id_mapping_text.encode(
            "utf-8"
        )
    ).hexdigest()
)


deduplicated_claims_write_status = (
    publish_frozen_artifact(
        deduplicated_claims_path,
        deduplicated_claims_text,
    )
)


claim_id_mapping_write_status = (
    publish_frozen_artifact(
        claim_id_mapping_path,
        claim_id_mapping_text,
    )
)


# =========================================================
# Deterministic report
# =========================================================

canonical_registry_report = {
    "report_version": (
        "claim_exact_canonicalization_report_v1"
    ),

    "canonical_registry_version": (
        CANONICAL_REGISTRY_VERSION
    ),

    "claim_id_version": (
        CLAIM_ID_VERSION
    ),

    "source_validated_claims": (
        validated_claims_path.name
    ),

    "source_validated_claims_sha256": (
        validated_claims_current_sha256
    ),

    "source_evidence_links": (
        full_evidence_path.name
    ),

    "source_evidence_links_sha256": (
        evidence_links_current_sha256
    ),

    "source_relation_ledger": (
        dedup_relation_ledger_path.name
    ),

    "source_relation_ledger_sha256": (
        relation_ledger_current_sha256
    ),

    "validated_raw_claims": (
        len(
            canonical_validated_claims
        )
    ),

    "canonical_claims": (
        len(
            deduplicated_claim_records
        )
    ),

    "claim_ids_assigned": (
        len(
            deduplicated_claim_records
        )
    ),

    "raw_claim_mapping_records": (
        len(
            claim_id_mapping_records
        )
    ),

    "exact_duplicate_canonical_groups": (
        exact_collapsed_canonical_claims
    ),

    "canonical_singletons": (
        singleton_canonical_claims
    ),

    "raw_claims_collapsed_by_exact_dedup": (
        raw_claims_collapsed_away
    ),

    "semantic_duplicate_merges": 0,

    "subsumption_merges": 0,

    "semantic_merge_applied": False,

    "subsumption_merge_applied": False,

    "validated_evidence_occurrences_preserved": (
        canonical_output_evidence_occurrences
    ),

    "metadata_multi_value_canonical_claims": dict(
        sorted(
            metadata_multi_value_counts.items()
        )
    ),

    "api_calls": 0,

    "next_stage": (
        "RELATION_CLAIM_ID_PROJECTION"
    ),

    "artifacts": {
        "deduplicated_claims": (
            deduplicated_claims_path.name
        ),

        "deduplicated_claims_sha256": (
            deduplicated_claims_sha256
        ),

        "claim_id_mapping": (
            claim_id_mapping_path.name
        ),

        "claim_id_mapping_sha256": (
            claim_id_mapping_sha256
        ),
    },
}


canonical_registry_report_text = (
    json.dumps(
        canonical_registry_report,
        ensure_ascii=False,
        indent=2,
        sort_keys=True,
    )
    + "\n"
)


canonical_registry_report_write_status = (
    publish_frozen_artifact(
        canonical_registry_report_path,
        canonical_registry_report_text,
    )
)


# =========================================================
# Independent disk revalidation
# =========================================================

assert (
    load_jsonl(
        deduplicated_claims_path
    )
    == deduplicated_claim_records
)


assert (
    load_jsonl(
        claim_id_mapping_path
    )
    == claim_id_mapping_records
)


assert (
    load_json(
        canonical_registry_report_path
    )
    == canonical_registry_report
)


assert (
    file_sha256(
        deduplicated_claims_path
    )
    == deduplicated_claims_sha256
)


assert (
    file_sha256(
        claim_id_mapping_path
    )
    == claim_id_mapping_sha256
)


# =========================================================
# Gate
# =========================================================

exact_canonical_registry_status = (
    "PASS"
)


relation_claim_id_projection_readiness = (
    "READY_FOR_RELATION_CLAIM_ID_PROJECTION"
)


# =========================================================
# Result
# =========================================================

print(
    "=== NB12i EXACT CANONICAL CLAIM REGISTRY ==="
)

print()

print(
    "Validated Raw Claims:",
    len(
        canonical_validated_claims
    ),
)

print(
    "Canonical Claims:",
    len(
        deduplicated_claim_records
    ),
)

print(
    "Claim IDs assigned:",
    len(
        deduplicated_claim_records
    ),
)


print()

print(
    "Exact Duplicate canonical groups:",
    exact_collapsed_canonical_claims,
)

print(
    "Canonical singletons:",
    singleton_canonical_claims,
)

print(
    "Raw Claims collapsed by Exact Dedup:",
    raw_claims_collapsed_away,
)


print()

print(
    "Semantic Duplicate merges:",
    0,
)

print(
    "Subsumption merges:",
    0,
)


print()

print(
    "Validated Evidence occurrences preserved:",
    canonical_output_evidence_occurrences,
)

print(
    "Metadata multi-value canonical Claims:",
    dict(
        sorted(
            metadata_multi_value_counts.items()
        )
    ),
)


print()

print(
    "Deduplicated Claims:",
    deduplicated_claims_path.relative_to(
        project_root
    ),
)

print(
    "Deduplicated Claims SHA-256:",
    deduplicated_claims_sha256,
)

print(
    "Write status:",
    deduplicated_claims_write_status,
)


print()

print(
    "Claim ID Mapping:",
    claim_id_mapping_path.relative_to(
        project_root
    ),
)

print(
    "Claim ID Mapping SHA-256:",
    claim_id_mapping_sha256,
)

print(
    "Mapping write status:",
    claim_id_mapping_write_status,
)


print()

print(
    "API calls performed:",
    0,
)

print(
    "Next stage:",
    "RELATION_CLAIM_ID_PROJECTION",
)

print(
    "Projection readiness:",
    relation_claim_id_projection_readiness,
)


print()

print(
    "CODE CELL 27 GATE:",
    exact_canonical_registry_status,
)
```

<div style="direction: rtl; white-space: normal; line-height: 1.7;">

<h2>Projection روابط به Claim IDهای نهایی</h2>

<p>
Canonical Claim Registry با موفقیت ساخته شده و ۴۰۰۶ Claim نهایی دارای
<code>claim_id</code> پایدار هستند.
در این مرحله Relationهای Dedup به این Claim IDهای نهایی متصل می‌شوند.
</p>

<p>
Exact Duplicateها پس از Canonicalization به یک Claim ID واحد می‌رسند؛
بنابراین به‌عنوان provenance نگهداری می‌شوند اما Edge واقعی Knowledge Graph محسوب نمی‌شوند.
روابط Semantic Duplicate، Subsumption و Partial Overlap بین Claimهای مستقل
به‌عنوان Edge قابل استفاده در Knowledge Graph باقی می‌مانند.
</p>

<ul>
<li>هیچ API یا مدل زبانی استفاده نمی‌شود.</li>
<li>تمام Relationهای قبلی حفظ می‌شوند.</li>
<li>Exact Duplicate provenance حذف نمی‌شود.</li>
<li>برای Subsumption، جهت Claim بالادست و پایین‌دست صریح ثبت می‌شود.</li>
<li>خروجی اصلی این مرحله <code>claim_relations_v1.jsonl</code> است.</li>
</ul>

</div>

```python
# Code Cell 28

assert (
    exact_canonical_registry_status
    == "PASS"
), (
    "Code Cell 27 must PASS before "
    "Relation Claim-ID Projection."
)

assert (
    relation_claim_id_projection_readiness
    == "READY_FOR_RELATION_CLAIM_ID_PROJECTION"
)


# =========================================================
# Configuration
# =========================================================

CLAIM_RELATION_PROJECTION_VERSION = (
    "claim_relation_projection_v1"
)


final_claim_relations_path = (
    registry_artifact_root
    / "claim_relations_v1.jsonl"
)

claim_relation_projection_report_path = (
    registry_artifact_root
    / "claim_relation_projection_report_v1.json"
)


# =========================================================
# Reload frozen inputs independently
# =========================================================

projection_claims = load_jsonl(
    deduplicated_claims_path
)

projection_mapping = load_jsonl(
    claim_id_mapping_path
)

projection_relation_ledger = load_jsonl(
    dedup_relation_ledger_path
)

projection_registry_report = load_json(
    canonical_registry_report_path
)


# =========================================================
# SHA-chain validation
# =========================================================

projection_claims_sha256 = (
    file_sha256(
        deduplicated_claims_path
    )
)

projection_mapping_sha256 = (
    file_sha256(
        claim_id_mapping_path
    )
)

projection_relation_ledger_sha256 = (
    file_sha256(
        dedup_relation_ledger_path
    )
)


assert (
    projection_claims_sha256
    == projection_registry_report[
        "artifacts"
    ][
        "deduplicated_claims_sha256"
    ]
), (
    "Deduplicated Claims SHA mismatch."
)


assert (
    projection_mapping_sha256
    == projection_registry_report[
        "artifacts"
    ][
        "claim_id_mapping_sha256"
    ]
), (
    "Claim-ID Mapping SHA mismatch."
)


assert (
    projection_relation_ledger_sha256
    == dedup_relation_ledger_sha256
), (
    "Dedup Relation Ledger SHA mismatch."
)


# =========================================================
# Canonical Claim index
# =========================================================

canonical_claim_by_id = {
    item[
        "claim_id"
    ]: item
    for item in (
        projection_claims
    )
}


assert (
    len(
        canonical_claim_by_id
    )
    == len(
        projection_claims
    )
)


assert (
    len(
        canonical_claim_by_id
    )
    == 4006
)


# =========================================================
# Raw Claim -> final Claim ID mapping
# =========================================================

raw_to_final_claim_id = {}


for item in (
    projection_mapping
):

    raw_claim_id = item[
        "raw_claim_id"
    ]

    claim_id = item[
        "claim_id"
    ]


    assert (
        raw_claim_id
        not in raw_to_final_claim_id
    )


    assert (
        claim_id
        in canonical_claim_by_id
    )


    raw_to_final_claim_id[
        raw_claim_id
    ] = claim_id


assert (
    len(
        raw_to_final_claim_id
    )
    == 4039
)


# =========================================================
# Projection helper
# =========================================================

def resolve_relation_side_claim_id(
    raw_claim_ids,
):

    assert (
        isinstance(
            raw_claim_ids,
            list,
        )
        and raw_claim_ids
    )


    resolved_claim_ids = {
        raw_to_final_claim_id[
            raw_claim_id
        ]
        for raw_claim_id in (
            raw_claim_ids
        )
    }


    assert (
        len(
            resolved_claim_ids
        )
        == 1
    ), (
        "A Relation side resolves to more than "
        "one canonical claim_id."
    )


    return next(
        iter(
            resolved_claim_ids
        )
    )


# =========================================================
# Project every Relation
# =========================================================

projected_claim_relations = []


for source_relation in (
    projection_relation_ledger
):

    relation_type = (
        source_relation[
            "relation_type"
        ]
    )


    assert (
        relation_type
        in ALLOWED_DEDUP_RELATION_TYPES
    )


    claim_id_a = (
        resolve_relation_side_claim_id(
            source_relation[
                "raw_claim_ids_a"
            ]
        )
    )


    claim_id_b = (
        resolve_relation_side_claim_id(
            source_relation[
                "raw_claim_ids_b"
            ]
        )
    )


    subsuming_claim_id = None
    subsumed_claim_id = None


    # -----------------------------------------------------
    # Exact duplicates collapse into the SAME canonical
    # Claim. Preserve provenance, but do not expose this as
    # a graph self-edge.
    # -----------------------------------------------------

    if (
        relation_type
        == "EXACT_DUPLICATE"
    ):

        assert (
            claim_id_a
            == claim_id_b
        ), (
            "EXACT_DUPLICATE members did not "
            "collapse to the same final claim_id."
        )


        assert (
            source_relation[
                "subsumption_direction"
            ]
            is None
        )


        graph_edge_eligible = False

        relation_role = (
            "CANONICALIZATION_PROVENANCE"
        )


    # -----------------------------------------------------
    # Semantic relations must connect different canonical
    # Claims because no automatic semantic merge occurred.
    # -----------------------------------------------------

    else:

        assert (
            claim_id_a
            != claim_id_b
        ), (
            "Semantic relation collapsed into a "
            "self-relation unexpectedly."
        )


        graph_edge_eligible = True

        relation_role = (
            "KNOWLEDGE_RELATION"
        )


        if (
            relation_type
            == "SUBSUMES"
        ):

            direction = (
                source_relation[
                    "subsumption_direction"
                ]
            )


            assert (
                direction
                in SUBSUMPTION_DIRECTIONS
            )


            if (
                direction
                == "A_SUBSUMES_B"
            ):

                subsuming_claim_id = (
                    claim_id_a
                )

                subsumed_claim_id = (
                    claim_id_b
                )


            elif (
                direction
                == "B_SUBSUMES_A"
            ):

                subsuming_claim_id = (
                    claim_id_b
                )

                subsumed_claim_id = (
                    claim_id_a
                )


        else:

            assert (
                source_relation[
                    "subsumption_direction"
                ]
                is None
            )


    # -----------------------------------------------------
    # Verify text/Claim-ID identity chain.
    # -----------------------------------------------------

    canonical_text_a = (
        canonical_claim_by_id[
            claim_id_a
        ][
            "claim_text_fa"
        ]
    )

    canonical_text_b = (
        canonical_claim_by_id[
            claim_id_b
        ][
            "claim_text_fa"
        ]
    )


    assert (
        canonical_text_a
        == source_relation[
            "claim_text_a"
        ]
    )


    assert (
        canonical_text_b
        == source_relation[
            "claim_text_b"
        ]
    )


    projected_claim_relations.append({
        "relation_id": (
            source_relation[
                "relation_id"
            ]
        ),

        "relation_type": (
            relation_type
        ),

        "relation_role": (
            relation_role
        ),

        "claim_id_a": (
            claim_id_a
        ),

        "claim_id_b": (
            claim_id_b
        ),

        "subsuming_claim_id": (
            subsuming_claim_id
        ),

        "subsumed_claim_id": (
            subsumed_claim_id
        ),

        "graph_edge_eligible": (
            graph_edge_eligible
        ),

        "source_relation_scope": (
            source_relation[
                "relation_scope"
            ]
        ),

        "candidate_pair_id": (
            source_relation[
                "candidate_pair_id"
            ]
        ),

        "raw_claim_ids_a": (
            source_relation[
                "raw_claim_ids_a"
            ]
        ),

        "raw_claim_ids_b": (
            source_relation[
                "raw_claim_ids_b"
            ]
        ),

        "claim_text_a": (
            source_relation[
                "claim_text_a"
            ]
        ),

        "claim_text_b": (
            source_relation[
                "claim_text_b"
            ]
        ),

        "cosine_similarity": (
            source_relation[
                "cosine_similarity"
            ]
        ),

        "subsumption_direction": (
            source_relation[
                "subsumption_direction"
            ]
        ),

        "judgment_origin": (
            source_relation[
                "judgment_origin"
            ]
        ),

        "source_stage": (
            source_relation[
                "source_stage"
            ]
        ),

        "projection_version": (
            CLAIM_RELATION_PROJECTION_VERSION
        ),
    })


# =========================================================
# Deterministic ordering
# =========================================================

projected_claim_relations = sorted(
    projected_claim_relations,
    key=lambda item: (
        item[
            "relation_type"
        ],
        item[
            "relation_id"
        ],
    ),
)


# =========================================================
# Structural validation
# =========================================================

projected_relation_ids = [
    item[
        "relation_id"
    ]
    for item in (
        projected_claim_relations
    )
]


assert (
    len(
        projected_relation_ids
    )
    == len(
        set(
            projected_relation_ids
        )
    )
)


assert (
    len(
        projected_claim_relations
    )
    == len(
        projection_relation_ledger
    )
)


assert (
    len(
        projected_claim_relations
    )
    == 2516
)


# =========================================================
# Relation diagnostics
# =========================================================

projected_relation_type_counts = Counter(
    item[
        "relation_type"
    ]
    for item in (
        projected_claim_relations
    )
)


graph_edge_relations = [
    item
    for item in (
        projected_claim_relations
    )
    if item[
        "graph_edge_eligible"
    ]
]


canonicalization_provenance_relations = [
    item
    for item in (
        projected_claim_relations
    )
    if not item[
        "graph_edge_eligible"
    ]
]


assert all(
    item[
        "relation_type"
    ]
    == "EXACT_DUPLICATE"
    for item in (
        canonicalization_provenance_relations
    )
)


assert all(
    item[
        "claim_id_a"
    ]
    == item[
        "claim_id_b"
    ]
    for item in (
        canonicalization_provenance_relations
    )
)


assert all(
    item[
        "claim_id_a"
    ]
    != item[
        "claim_id_b"
    ]
    for item in (
        graph_edge_relations
    )
)


# =========================================================
# Validate SUBSUMES projection
# =========================================================

subsumes_projected = [
    item
    for item in (
        projected_claim_relations
    )
    if (
        item[
            "relation_type"
        ]
        == "SUBSUMES"
    )
]


for item in (
    subsumes_projected
):

    assert (
        item[
            "subsuming_claim_id"
        ]
        in {
            item[
                "claim_id_a"
            ],
            item[
                "claim_id_b"
            ],
        }
    )


    assert (
        item[
            "subsumed_claim_id"
        ]
        in {
            item[
                "claim_id_a"
            ],
            item[
                "claim_id_b"
            ],
        }
    )


    assert (
        item[
            "subsuming_claim_id"
        ]
        != item[
            "subsumed_claim_id"
        ]
    )


# =========================================================
# Check canonical graph-edge uniqueness
# =========================================================

canonical_graph_edge_keys = []


for item in (
    graph_edge_relations
):

    relation_type = (
        item[
            "relation_type"
        ]
    )


    if (
        relation_type
        == "SUBSUMES"
    ):

        edge_key = (
            relation_type,
            item[
                "subsuming_claim_id"
            ],
            item[
                "subsumed_claim_id"
            ],
        )


    else:

        ordered_claim_ids = tuple(
            sorted([
                item[
                    "claim_id_a"
                ],
                item[
                    "claim_id_b"
                ],
            ])
        )


        edge_key = (
            relation_type,
            ordered_claim_ids[0],
            ordered_claim_ids[1],
        )


    canonical_graph_edge_keys.append(
        edge_key
    )


assert (
    len(
        canonical_graph_edge_keys
    )
    == len(
        set(
            canonical_graph_edge_keys
        )
    )
), (
    "Duplicate canonical graph relation detected."
)


# =========================================================
# Serialize final Claim Relation artifact
# =========================================================

final_claim_relations_text = "".join(
    json.dumps(
        item,
        ensure_ascii=False,
        sort_keys=True,
    )
    + "\n"
    for item in (
        projected_claim_relations
    )
)


final_claim_relations_sha256 = (
    hashlib.sha256(
        final_claim_relations_text.encode(
            "utf-8"
        )
    ).hexdigest()
)


final_claim_relations_write_status = (
    publish_frozen_artifact(
        final_claim_relations_path,
        final_claim_relations_text,
    )
)


# =========================================================
# Deterministic projection report
# =========================================================

claim_relation_projection_report = {
    "report_version": (
        "claim_relation_projection_report_v1"
    ),

    "projection_version": (
        CLAIM_RELATION_PROJECTION_VERSION
    ),

    "source_deduplicated_claims": (
        deduplicated_claims_path.name
    ),

    "source_deduplicated_claims_sha256": (
        projection_claims_sha256
    ),

    "source_claim_id_mapping": (
        claim_id_mapping_path.name
    ),

    "source_claim_id_mapping_sha256": (
        projection_mapping_sha256
    ),

    "source_relation_ledger": (
        dedup_relation_ledger_path.name
    ),

    "source_relation_ledger_sha256": (
        projection_relation_ledger_sha256
    ),

    "canonical_claims": (
        len(
            projection_claims
        )
    ),

    "projected_relation_records": (
        len(
            projected_claim_relations
        )
    ),

    "relation_type_counts": dict(
        sorted(
            projected_relation_type_counts.items()
        )
    ),

    "graph_edge_eligible_relations": (
        len(
            graph_edge_relations
        )
    ),

    "canonicalization_provenance_relations": (
        len(
            canonicalization_provenance_relations
        )
    ),

    "exact_duplicate_self_relations": (
        len(
            canonicalization_provenance_relations
        )
    ),

    "semantic_self_relations": 0,

    "subsumes_relations_with_explicit_direction": (
        len(
            subsumes_projected
        )
    ),

    "automatic_merge_applied": False,

    "claim_ids_reassigned": False,

    "api_calls": 0,

    "next_stage": (
        "KNOWLEDGE_GRAPH_BUILD"
    ),

    "artifact": {
        "claim_relations": (
            final_claim_relations_path.name
        ),

        "claim_relations_sha256": (
            final_claim_relations_sha256
        ),
    },
}


claim_relation_projection_report_text = (
    json.dumps(
        claim_relation_projection_report,
        ensure_ascii=False,
        indent=2,
        sort_keys=True,
    )
    + "\n"
)


claim_relation_projection_report_write_status = (
    publish_frozen_artifact(
        claim_relation_projection_report_path,
        claim_relation_projection_report_text,
    )
)


# =========================================================
# Independent disk revalidation
# =========================================================

assert (
    load_jsonl(
        final_claim_relations_path
    )
    == projected_claim_relations
)


assert (
    load_json(
        claim_relation_projection_report_path
    )
    == claim_relation_projection_report
)


assert (
    file_sha256(
        final_claim_relations_path
    )
    == final_claim_relations_sha256
)


assert (
    claim_relation_projection_report[
        "artifact"
    ][
        "claim_relations_sha256"
    ]
    == final_claim_relations_sha256
)


# =========================================================
# Gate
# =========================================================

claim_relation_projection_status = (
    "PASS"
)


knowledge_graph_build_readiness = (
    "READY_FOR_KNOWLEDGE_GRAPH_BUILD"
)


# =========================================================
# Result
# =========================================================

print(
    "=== NB12i RELATION CLAIM-ID PROJECTION ==="
)

print()

print(
    "Canonical Claims:",
    len(
        projection_claims
    ),
)

print(
    "Input Relation Ledger records:",
    len(
        projection_relation_ledger
    ),
)

print(
    "Projected Claim Relations:",
    len(
        projected_claim_relations
    ),
)


print()

print(
    "Relation type counts:",
    dict(
        sorted(
            projected_relation_type_counts.items()
        )
    ),
)

print(
    "Graph-edge eligible relations:",
    len(
        graph_edge_relations
    ),
)

print(
    "Exact canonicalization provenance relations:",
    len(
        canonicalization_provenance_relations
    ),
)

print(
    "Semantic self-relations:",
    0,
)

print(
    "SUBSUMES relations with explicit direction:",
    len(
        subsumes_projected
    ),
)


print()

print(
    "Claim Relations:",
    final_claim_relations_path.relative_to(
        project_root
    ),
)

print(
    "Claim Relations SHA-256:",
    final_claim_relations_sha256,
)

print(
    "Relation write status:",
    final_claim_relations_write_status,
)

print(
    "Report write status:",
    claim_relation_projection_report_write_status,
)


print()

print(
    "API calls performed:",
    0,
)

print(
    "Next stage:",
    "KNOWLEDGE_GRAPH_BUILD",
)

print(
    "Knowledge Graph readiness:",
    knowledge_graph_build_readiness,
)


print()

print(
    "CODE CELL 28 GATE:",
    claim_relation_projection_status,
)
```

در این مرحله Knowledge Graph نسخه اول فقط از اطلاعاتی ساخته می‌شود
که قبلاً به‌صورت معتبر و قابل ردیابی تولید شده‌اند.

استنتاج یا تولید نمی‌شود.

<li>Source identity و Document lineage حفظ می‌شود.</li>
<li>۲۴۱۵ Relation قابل استفاده بین Claimها وارد Graph می‌شوند.</li>
<li>۱۰۱ Exact Duplicate Relation به‌عنوان Canonicalization Provenance نگهداری می‌شوند و Graph self-edge ایجاد نمی‌کنند.</li>
<li>هیچ API یا مدل زبانی استفاده نمی‌شود.</li>
</ul>

```python
# Code Cell 29

assert (
    claim_relation_projection_status
    == "PASS"
), (
    "Code Cell 28 must PASS before "
    "Knowledge Graph Build."
)

assert (
    knowledge_graph_build_readiness
    == "READY_FOR_KNOWLEDGE_GRAPH_BUILD"
)


# =========================================================
# Configuration
# =========================================================

KNOWLEDGE_GRAPH_VERSION = (
    "knowledge_graph_v1"
)

KNOWLEDGE_GRAPH_SCHEMA_VERSION = (
    "claim_evidence_source_graph_v1"
)


knowledge_graph_path = (
    registry_artifact_root
    / "knowledge_graph_v1.json"
)

knowledge_graph_report_path = (
    registry_artifact_root
    / "knowledge_graph_build_report_v1.json"
)


# =========================================================
# Independently reload frozen inputs
# =========================================================

kg_claims = load_jsonl(
    deduplicated_claims_path
)

kg_raw_to_claim_mapping = load_jsonl(
    claim_id_mapping_path
)

kg_evidence_links = load_jsonl(
    full_evidence_path
)

kg_claim_relations = load_jsonl(
    final_claim_relations_path
)

kg_relation_projection_report = load_json(
    claim_relation_projection_report_path
)


# =========================================================
# SHA-chain validation
# =========================================================

kg_claims_sha256 = file_sha256(
    deduplicated_claims_path
)

kg_mapping_sha256 = file_sha256(
    claim_id_mapping_path
)

kg_evidence_sha256 = file_sha256(
    full_evidence_path
)

kg_relations_sha256 = file_sha256(
    final_claim_relations_path
)


assert (
    kg_claims_sha256
    == deduplicated_claims_sha256
), (
    "Deduplicated Claim artifact changed."
)


assert (
    kg_mapping_sha256
    == claim_id_mapping_sha256
), (
    "Claim-ID Mapping artifact changed."
)


assert (
    kg_relations_sha256
    == final_claim_relations_sha256
), (
    "Claim Relation artifact changed."
)


assert (
    kg_relation_projection_report[
        "artifact"
    ][
        "claim_relations_sha256"
    ]
    == kg_relations_sha256
)


# =========================================================
# Claim index
# =========================================================

kg_claim_by_id = {
    item[
        "claim_id"
    ]: item
    for item in (
        kg_claims
    )
}


assert (
    len(
        kg_claim_by_id
    )
    == len(
        kg_claims
    )
)


assert (
    len(
        kg_claim_by_id
    )
    == 4006
)


# =========================================================
# Raw Claim -> final Claim mapping
# =========================================================

kg_raw_to_claim_id = {}


for item in (
    kg_raw_to_claim_mapping
):

    raw_claim_id = item[
        "raw_claim_id"
    ]

    claim_id = item[
        "claim_id"
    ]


    assert (
        raw_claim_id
        not in kg_raw_to_claim_id
    )


    assert (
        claim_id
        in kg_claim_by_id
    )


    kg_raw_to_claim_id[
        raw_claim_id
    ] = claim_id


assert (
    len(
        kg_raw_to_claim_id
    )
    == 4039
)


# =========================================================
# Restrict Evidence to final VALIDATED Registry Claims
# =========================================================

kg_validated_evidence = [
    evidence
    for evidence in (
        kg_evidence_links
    )
    if (
        evidence[
            "raw_claim_id"
        ]
        in kg_raw_to_claim_id
    )
]


kg_evidence_by_id = {}


for evidence in (
    kg_validated_evidence
):

    evidence_id = evidence[
        "evidence_id"
    ]


    assert (
        evidence_id
        not in kg_evidence_by_id
    ), (
        "Duplicate evidence_id detected: "
        f"{evidence_id}"
    )


    kg_evidence_by_id[
        evidence_id
    ] = evidence


assert (
    len(
        kg_evidence_by_id
    )
    == len(
        kg_validated_evidence
    )
)


assert (
    len(
        kg_validated_evidence
    )
    == 4125
), (
    "Validated Evidence occurrence count "
    "differs from Code Cell 27."
)


# =========================================================
# Build Claim Nodes
# =========================================================

kg_claim_nodes = []


for claim in (
    kg_claims
):

    kg_claim_nodes.append({
        "node_id": (
            claim[
                "claim_id"
            ]
        ),

        "node_type": (
            "CLAIM"
        ),

        "claim_id": (
            claim[
                "claim_id"
            ]
        ),

        "claim_text_fa": (
            claim[
                "claim_text_fa"
            ]
        ),

        "claim_text_sha256": (
            claim[
                "claim_text_sha256"
            ]
        ),

        "claim_type": (
            claim[
                "claim_type"
            ]
        ),

        "claim_types": (
            claim[
                "claim_types"
            ]
        ),

        "authority_class": (
            claim[
                "authority_class"
            ]
        ),

        "authority_classes": (
            claim[
                "authority_classes"
            ]
        ),

        "claim_discovery_role": (
            claim[
                "claim_discovery_role"
            ]
        ),

        "claim_discovery_roles": (
            claim[
                "claim_discovery_roles"
            ]
        ),

        "source_types": (
            claim[
                "source_types"
            ]
        ),

        "source_ids": (
            claim[
                "source_ids"
            ]
        ),

        "source_document_ids": (
            claim[
                "source_document_ids"
            ]
        ),

        "evidence_ids": (
            claim[
                "evidence_ids"
            ]
        ),

        "evidence_count": (
            claim[
                "evidence_count"
            ]
        ),

        "raw_claim_ids": (
            claim[
                "raw_claim_ids"
            ]
        ),

        "raw_claim_count": (
            claim[
                "raw_claim_count"
            ]
        ),

        "exact_duplicate_collapsed": (
            claim[
                "exact_duplicate_collapsed"
            ]
        ),

        "registry_validation_status": (
            claim[
                "registry_validation_status"
            ]
        ),
    })


kg_claim_nodes = sorted(
    kg_claim_nodes,
    key=lambda item: (
        item[
            "node_id"
        ]
    ),
)


# =========================================================
# Build Evidence Nodes
# =========================================================

kg_evidence_nodes = []


for evidence in (
    kg_validated_evidence
):

    raw_claim_id = evidence[
        "raw_claim_id"
    ]

    claim_id = (
        kg_raw_to_claim_id[
            raw_claim_id
        ]
    )


    kg_evidence_nodes.append({
        "node_id": (
            "evidence:"
            + evidence[
                "evidence_id"
            ]
        ),

        "node_type": (
            "EVIDENCE"
        ),

        "evidence_id": (
            evidence[
                "evidence_id"
            ]
        ),

        "claim_id": (
            claim_id
        ),

        "raw_claim_id": (
            raw_claim_id
        ),

        "document_id": (
            evidence[
                "document_id"
            ]
        ),

        "source_id": (
            evidence[
                "source_id"
            ]
        ),

        "source_type": (
            evidence[
                "source_type"
            ]
        ),

        "authority_class": (
            evidence[
                "authority_class"
            ]
        ),

        "claim_discovery_role": (
            evidence[
                "claim_discovery_role"
            ]
        ),

        "evidence_quote_exact": (
            evidence[
                "evidence_quote_exact"
            ]
        ),

        "locator": (
            evidence[
                "locator"
            ]
        ),

        "attribution_snapshot": (
            evidence[
                "attribution_snapshot"
            ]
        ),

        "evidence_text_sha256": (
            evidence[
                "evidence_text_sha256"
            ]
        ),
    })


kg_evidence_nodes = sorted(
    kg_evidence_nodes,
    key=lambda item: (
        item[
            "evidence_id"
        ]
    ),
)


# =========================================================
# Build Source Nodes
#
# A Source node represents source_id.
# All observed document IDs and source types are preserved.
# =========================================================

source_documents = defaultdict(
    set
)

source_types = defaultdict(
    set
)

source_authority_classes = defaultdict(
    set
)

source_discovery_roles = defaultdict(
    set
)


for evidence in (
    kg_validated_evidence
):

    source_id = evidence[
        "source_id"
    ]


    source_documents[
        source_id
    ].add(
        evidence[
            "document_id"
        ]
    )


    source_types[
        source_id
    ].add(
        evidence[
            "source_type"
        ]
    )


    source_authority_classes[
        source_id
    ].add(
        evidence[
            "authority_class"
        ]
    )


    source_discovery_roles[
        source_id
    ].add(
        evidence[
            "claim_discovery_role"
        ]
    )


kg_source_nodes = []


for source_id in sorted(
    source_documents
):

    kg_source_nodes.append({
        "node_id": (
            "source:"
            + source_id
        ),

        "node_type": (
            "SOURCE"
        ),

        "source_id": (
            source_id
        ),

        "document_ids": sorted(
            source_documents[
                source_id
            ]
        ),

        "source_types": sorted(
            source_types[
                source_id
            ]
        ),

        "authority_classes": sorted(
            source_authority_classes[
                source_id
            ]
        ),

        "claim_discovery_roles": sorted(
            source_discovery_roles[
                source_id
            ]
        ),
    })


# =========================================================
# Claim -> Evidence structural edges
# =========================================================

kg_claim_evidence_edges = []


for evidence in (
    kg_validated_evidence
):

    claim_id = (
        kg_raw_to_claim_id[
            evidence[
                "raw_claim_id"
            ]
        ]
    )


    edge_material = (
        "CLAIM_HAS_EVIDENCE"
        + "\n"
        + claim_id
        + "\n"
        + evidence[
            "evidence_id"
        ]
    )


    edge_sha256 = hashlib.sha256(
        edge_material.encode(
            "utf-8"
        )
    ).hexdigest()


    kg_claim_evidence_edges.append({
        "edge_id": (
            "graph-edge:"
            + edge_sha256[:24]
        ),

        "edge_type": (
            "CLAIM_HAS_EVIDENCE"
        ),

        "from_node_id": (
            claim_id
        ),

        "to_node_id": (
            "evidence:"
            + evidence[
                "evidence_id"
            ]
        ),

        "claim_id": (
            claim_id
        ),

        "evidence_id": (
            evidence[
                "evidence_id"
            ]
        ),
    })


kg_claim_evidence_edges = sorted(
    kg_claim_evidence_edges,
    key=lambda item: (
        item[
            "edge_id"
        ]
    ),
)


# =========================================================
# Evidence -> Source structural edges
# =========================================================

kg_evidence_source_edges = []


for evidence in (
    kg_validated_evidence
):

    edge_material = (
        "EVIDENCE_FROM_SOURCE"
        + "\n"
        + evidence[
            "evidence_id"
        ]
        + "\n"
        + evidence[
            "source_id"
        ]
    )


    edge_sha256 = hashlib.sha256(
        edge_material.encode(
            "utf-8"
        )
    ).hexdigest()


    kg_evidence_source_edges.append({
        "edge_id": (
            "graph-edge:"
            + edge_sha256[:24]
        ),

        "edge_type": (
            "EVIDENCE_FROM_SOURCE"
        ),

        "from_node_id": (
            "evidence:"
            + evidence[
                "evidence_id"
            ]
        ),

        "to_node_id": (
            "source:"
            + evidence[
                "source_id"
            ]
        ),

        "evidence_id": (
            evidence[
                "evidence_id"
            ]
        ),

        "source_id": (
            evidence[
                "source_id"
            ]
        ),

        "document_id": (
            evidence[
                "document_id"
            ]
        ),
    })


kg_evidence_source_edges = sorted(
    kg_evidence_source_edges,
    key=lambda item: (
        item[
            "edge_id"
        ]
    ),
)


# =========================================================
# Claim -> Claim semantic/dedup relation edges
#
# Only graph_edge_eligible=True relations become graph edges.
# EXACT_DUPLICATE provenance is retained separately.
# =========================================================

kg_claim_relation_edges = []

kg_exact_duplicate_provenance = []


for relation in (
    kg_claim_relations
):

    if not relation[
        "graph_edge_eligible"
    ]:

        assert (
            relation[
                "relation_type"
            ]
            == "EXACT_DUPLICATE"
        )


        kg_exact_duplicate_provenance.append({
            "relation_id": (
                relation[
                    "relation_id"
                ]
            ),

            "relation_type": (
                relation[
                    "relation_type"
                ]
            ),

            "claim_id": (
                relation[
                    "claim_id_a"
                ]
            ),

            "raw_claim_ids_a": (
                relation[
                    "raw_claim_ids_a"
                ]
            ),

            "raw_claim_ids_b": (
                relation[
                    "raw_claim_ids_b"
                ]
            ),

            "source_stage": (
                relation[
                    "source_stage"
                ]
            ),
        })


        continue


    claim_id_a = relation[
        "claim_id_a"
    ]

    claim_id_b = relation[
        "claim_id_b"
    ]


    assert (
        claim_id_a
        in kg_claim_by_id
    )

    assert (
        claim_id_b
        in kg_claim_by_id
    )

    assert (
        claim_id_a
        != claim_id_b
    )


    relation_type = relation[
        "relation_type"
    ]


    if (
        relation_type
        == "SUBSUMES"
    ):

        from_claim_id = (
            relation[
                "subsuming_claim_id"
            ]
        )

        to_claim_id = (
            relation[
                "subsumed_claim_id"
            ]
        )

        directed = True


    else:

        ordered_ids = sorted([
            claim_id_a,
            claim_id_b,
        ])

        from_claim_id = (
            ordered_ids[0]
        )

        to_claim_id = (
            ordered_ids[1]
        )

        directed = False


    kg_claim_relation_edges.append({
        "edge_id": (
            relation[
                "relation_id"
            ]
        ),

        "edge_type": (
            relation_type
        ),

        "from_node_id": (
            from_claim_id
        ),

        "to_node_id": (
            to_claim_id
        ),

        "directed": (
            directed
        ),

        "claim_id_a": (
            claim_id_a
        ),

        "claim_id_b": (
            claim_id_b
        ),

        "subsuming_claim_id": (
            relation[
                "subsuming_claim_id"
            ]
        ),

        "subsumed_claim_id": (
            relation[
                "subsumed_claim_id"
            ]
        ),

        "cosine_similarity": (
            relation[
                "cosine_similarity"
            ]
        ),

        "judgment_origin": (
            relation[
                "judgment_origin"
            ]
        ),

        "candidate_pair_id": (
            relation[
                "candidate_pair_id"
            ]
        ),
    })


kg_claim_relation_edges = sorted(
    kg_claim_relation_edges,
    key=lambda item: (
        item[
            "edge_id"
        ]
    ),
)


kg_exact_duplicate_provenance = sorted(
    kg_exact_duplicate_provenance,
    key=lambda item: (
        item[
            "relation_id"
        ]
    ),
)


# =========================================================
# Graph consistency validation
# =========================================================

assert (
    len(
        kg_claim_relation_edges
    )
    == 2415
)


assert (
    len(
        kg_exact_duplicate_provenance
    )
    == 101
)


assert (
    len(
        kg_claim_evidence_edges
    )
    == len(
        kg_validated_evidence
    )
)


assert (
    len(
        kg_evidence_source_edges
    )
    == len(
        kg_validated_evidence
    )
)


# ---------------------------------------------------------
# All node IDs must be globally unique.
# ---------------------------------------------------------

all_node_ids = (
    [
        item[
            "node_id"
        ]
        for item in (
            kg_claim_nodes
        )
    ]
    + [
        item[
            "node_id"
        ]
        for item in (
            kg_evidence_nodes
        )
    ]
    + [
        item[
            "node_id"
        ]
        for item in (
            kg_source_nodes
        )
    ]
)


assert (
    len(
        all_node_ids
    )
    == len(
        set(
            all_node_ids
        )
    )
), (
    "Duplicate Knowledge Graph node_id detected."
)


all_node_id_set = set(
    all_node_ids
)


# ---------------------------------------------------------
# All structural/relation edge endpoints must exist.
# ---------------------------------------------------------

all_graph_edges = (
    kg_claim_relation_edges
    + kg_claim_evidence_edges
    + kg_evidence_source_edges
)


all_edge_ids = [
    edge[
        "edge_id"
    ]
    for edge in (
        all_graph_edges
    )
]


assert (
    len(
        all_edge_ids
    )
    == len(
        set(
            all_edge_ids
        )
    )
), (
    "Duplicate Knowledge Graph edge_id detected."
)


for edge in (
    all_graph_edges
):

    assert (
        edge[
            "from_node_id"
        ]
        in all_node_id_set
    )

    assert (
        edge[
            "to_node_id"
        ]
        in all_node_id_set
    )


# =========================================================
# Claim coverage validation
# =========================================================

claim_ids_with_evidence_edges = {
    edge[
        "claim_id"
    ]
    for edge in (
        kg_claim_evidence_edges
    )
}


assert (
    claim_ids_with_evidence_edges
    == set(
        kg_claim_by_id
    )
), (
    "At least one canonical Claim has no "
    "Evidence edge in Knowledge Graph."
)


# =========================================================
# Evidence identity validation
# =========================================================

evidence_node_ids = {
    item[
        "evidence_id"
    ]
    for item in (
        kg_evidence_nodes
    )
}


assert (
    evidence_node_ids
    == {
        evidence[
            "evidence_id"
        ]
        for evidence in (
            kg_validated_evidence
        )
    }
)


# =========================================================
# Relation-type diagnostics
# =========================================================

kg_claim_relation_type_counts = Counter(
    edge[
        "edge_type"
    ]
    for edge in (
        kg_claim_relation_edges
    )
)


assert (
    kg_claim_relation_type_counts[
        "SEMANTIC_DUPLICATE"
    ]
    == 233
)


assert (
    kg_claim_relation_type_counts[
        "SUBSUMES"
    ]
    == 544
)


assert (
    kg_claim_relation_type_counts[
        "PARTIAL_OVERLAP"
    ]
    == 1638
)


# =========================================================
# IMPORTANT CONTRACT BOUNDARY
#
# The contract lists future knowledge relation classes such
# as RELATED_TO / SUPPORTS / CONTRADICTS / PREREQUISITE_OF.
#
# These have NOT been separately adjudicated in NB12i.
# Therefore none are fabricated here.
# =========================================================

generated_knowledge_relation_types = []

declared_future_knowledge_relation_types = (
    claim_registry_contract[
        "knowledge_relation_types"
    ]
)


# =========================================================
# Build deterministic Knowledge Graph object
# =========================================================

knowledge_graph = {
    "graph_version": (
        KNOWLEDGE_GRAPH_VERSION
    ),

    "schema_version": (
        KNOWLEDGE_GRAPH_SCHEMA_VERSION
    ),

    "graph_role": (
        "EVIDENCE_GROUNDED_CLAIM_REGISTRY_GRAPH"
    ),

    "source_artifacts": {
        "claims": {
            "path": (
                deduplicated_claims_path.name
            ),

            "sha256": (
                kg_claims_sha256
            ),
        },

        "claim_id_mapping": {
            "path": (
                claim_id_mapping_path.name
            ),

            "sha256": (
                kg_mapping_sha256
            ),
        },

        "evidence_links": {
            "path": (
                full_evidence_path.name
            ),

            "sha256": (
                kg_evidence_sha256
            ),
        },

        "claim_relations": {
            "path": (
                final_claim_relations_path.name
            ),

            "sha256": (
                kg_relations_sha256
            ),
        },
    },

    "nodes": {
        "claims": (
            kg_claim_nodes
        ),

        "evidence": (
            kg_evidence_nodes
        ),

        "sources": (
            kg_source_nodes
        ),
    },

    "edges": {
        "claim_relations": (
            kg_claim_relation_edges
        ),

        "claim_evidence": (
            kg_claim_evidence_edges
        ),

        "evidence_source": (
            kg_evidence_source_edges
        ),
    },

    "canonicalization_provenance": {
        "exact_duplicate_relations": (
            kg_exact_duplicate_provenance
        ),
    },

    "relation_scope": {
        "generated_claim_relation_types": sorted(
            kg_claim_relation_type_counts
        ),

        "generated_knowledge_relation_types": (
            generated_knowledge_relation_types
        ),

        "declared_future_knowledge_relation_types": (
            declared_future_knowledge_relation_types
        ),

        "future_knowledge_relations_inferred": False,
    },

    "statistics": {
        "claim_nodes": (
            len(
                kg_claim_nodes
            )
        ),

        "evidence_nodes": (
            len(
                kg_evidence_nodes
            )
        ),

        "source_nodes": (
            len(
                kg_source_nodes
            )
        ),

        "claim_relation_edges": (
            len(
                kg_claim_relation_edges
            )
        ),

        "claim_evidence_edges": (
            len(
                kg_claim_evidence_edges
            )
        ),

        "evidence_source_edges": (
            len(
                kg_evidence_source_edges
            )
        ),

        "exact_duplicate_provenance_relations": (
            len(
                kg_exact_duplicate_provenance
            )
        ),

        "claim_relation_type_counts": dict(
            sorted(
                kg_claim_relation_type_counts.items()
            )
        ),
    },

    "policies": {
        "automatic_semantic_merge_applied": False,
        "automatic_subsumption_merge_applied": False,
        "original_source_remains_final_evidence": True,
        "retrieval_chunks_remain_separate": True,
        "api_calls": 0,
    },
}


# =========================================================
# Serialize frozen Graph
# =========================================================

knowledge_graph_text = (
    json.dumps(
        knowledge_graph,
        ensure_ascii=False,
        indent=2,
        sort_keys=True,
    )
    + "\n"
)


knowledge_graph_sha256 = (
    hashlib.sha256(
        knowledge_graph_text.encode(
            "utf-8"
        )
    ).hexdigest()
)


knowledge_graph_write_status = (
    publish_frozen_artifact(
        knowledge_graph_path,
        knowledge_graph_text,
    )
)


# =========================================================
# Deterministic Graph report
# =========================================================

knowledge_graph_report = {
    "report_version": (
        "knowledge_graph_build_report_v1"
    ),

    "graph_version": (
        KNOWLEDGE_GRAPH_VERSION
    ),

    "schema_version": (
        KNOWLEDGE_GRAPH_SCHEMA_VERSION
    ),

    "claim_nodes": (
        len(
            kg_claim_nodes
        )
    ),

    "evidence_nodes": (
        len(
            kg_evidence_nodes
        )
    ),

    "source_nodes": (
        len(
            kg_source_nodes
        )
    ),

    "claim_relation_edges": (
        len(
            kg_claim_relation_edges
        )
    ),

    "claim_evidence_edges": (
        len(
            kg_claim_evidence_edges
        )
    ),

    "evidence_source_edges": (
        len(
            kg_evidence_source_edges
        )
    ),

    "exact_duplicate_provenance_relations": (
        len(
            kg_exact_duplicate_provenance
        )
    ),

    "claim_relation_type_counts": dict(
        sorted(
            kg_claim_relation_type_counts.items()
        )
    ),

    "canonical_claims_without_evidence": 0,

    "semantic_self_edges": 0,

    "future_knowledge_relations_inferred": False,

    "generated_knowledge_relation_types": [],

    "automatic_merge_applied": False,

    "api_calls": 0,

    "next_stage": (
        "FINAL_CLAIM_REGISTRY_REPORT"
    ),

    "artifact": {
        "knowledge_graph": (
            knowledge_graph_path.name
        ),

        "knowledge_graph_sha256": (
            knowledge_graph_sha256
        ),
    },
}


knowledge_graph_report_text = (
    json.dumps(
        knowledge_graph_report,
        ensure_ascii=False,
        indent=2,
        sort_keys=True,
    )
    + "\n"
)


knowledge_graph_report_write_status = (
    publish_frozen_artifact(
        knowledge_graph_report_path,
        knowledge_graph_report_text,
    )
)


# =========================================================
# Independent disk revalidation
# =========================================================

assert (
    load_json(
        knowledge_graph_path
    )
    == knowledge_graph
)


assert (
    load_json(
        knowledge_graph_report_path
    )
    == knowledge_graph_report
)


assert (
    file_sha256(
        knowledge_graph_path
    )
    == knowledge_graph_sha256
)


assert (
    knowledge_graph_report[
        "artifact"
    ][
        "knowledge_graph_sha256"
    ]
    == knowledge_graph_sha256
)


# =========================================================
# Gate
# =========================================================

knowledge_graph_build_status = (
    "PASS"
)


final_registry_report_readiness = (
    "READY_FOR_FINAL_CLAIM_REGISTRY_REPORT"
)


# =========================================================
# Result
# =========================================================

print(
    "=== NB12i KNOWLEDGE GRAPH BUILD ==="
)

print()

print(
    "Claim Nodes:",
    len(
        kg_claim_nodes
    ),
)

print(
    "Evidence Nodes:",
    len(
        kg_evidence_nodes
    ),
)

print(
    "Source Nodes:",
    len(
        kg_source_nodes
    ),
)


print()

print(
    "Claim Relation Edges:",
    len(
        kg_claim_relation_edges
    ),
)

print(
    "Claim -> Evidence Edges:",
    len(
        kg_claim_evidence_edges
    ),
)

print(
    "Evidence -> Source Edges:",
    len(
        kg_evidence_source_edges
    ),
)

print(
    "Exact Duplicate provenance relations:",
    len(
        kg_exact_duplicate_provenance
    ),
)


print()

print(
    "Claim Relation type counts:",
    dict(
        sorted(
            kg_claim_relation_type_counts.items()
        )
    ),
)


print()

print(
    "Canonical Claims without Evidence:",
    0,
)

print(
    "Semantic self-edges:",
    0,
)

print(
    "Future knowledge relations inferred:",
    False,
)

print(
    "API calls performed:",
    0,
)


print()

print(
    "Knowledge Graph:",
    knowledge_graph_path.relative_to(
        project_root
    ),
)

print(
    "Knowledge Graph SHA-256:",
    knowledge_graph_sha256,
)

print(
    "Graph write status:",
    knowledge_graph_write_status,
)

print(
    "Report write status:",
    knowledge_graph_report_write_status,
)


print()

print(
    "Next stage:",
    "FINAL_CLAIM_REGISTRY_REPORT",
)

print(
    "Registry report readiness:",
    final_registry_report_readiness,
)


print()

print(
    "CODE CELL 29 GATE:",
    knowledge_graph_build_status,
)
```

```python
# Code Cell 30

assert (
    knowledge_graph_build_status
    == "PASS"
), (
    "Code Cell 29 must PASS before "
    "Final Claim Registry Report."
)

assert (
    final_registry_report_readiness
    == "READY_FOR_FINAL_CLAIM_REGISTRY_REPORT"
)


# =========================================================
# Configuration
# =========================================================

FINAL_REGISTRY_REPORT_VERSION = (
    "claim_registry_report_v1"
)

FINAL_REGISTRY_STATUS = (
    "PASS"
)


final_registry_report_path = (
    registry_artifact_root
    / "claim_registry_report_v1.json"
)


# =========================================================
# Required Registry artifacts
# =========================================================

final_registry_required_paths = {
    "contract": (
        registry_artifact_root
        / "claim_registry_contract_v1.json"
    ),

    "raw_claims": (
        registry_artifact_root
        / "claims_raw_v1.jsonl"
    ),

    "validated_claims": (
        registry_artifact_root
        / "claims_validated_v1.jsonl"
    ),

    "deduplicated_claims": (
        registry_artifact_root
        / "claims_deduplicated_v1.jsonl"
    ),

    "evidence_links": (
        registry_artifact_root
        / "claim_evidence_links_v1.jsonl"
    ),

    "claim_relations": (
        registry_artifact_root
        / "claim_relations_v1.jsonl"
    ),

    "knowledge_graph": (
        registry_artifact_root
        / "knowledge_graph_v1.json"
    ),

    # Supplemental lineage artifact created after
    # the original frozen artifact-family contract.
    "claim_id_mapping": (
        registry_artifact_root
        / "claim_id_mapping_v1.jsonl"
    ),
}


for (
    artifact_name,
    artifact_path,
) in (
    final_registry_required_paths.items()
):

    assert (
        artifact_path.exists()
    ), (
        "Required Registry artifact missing: "
        f"{artifact_name} -> {artifact_path}"
    )


# =========================================================
# Independently reload everything from disk
# =========================================================

final_contract = load_json(
    final_registry_required_paths[
        "contract"
    ]
)

final_raw_claims = load_jsonl(
    final_registry_required_paths[
        "raw_claims"
    ]
)

final_validated_claims = load_jsonl(
    final_registry_required_paths[
        "validated_claims"
    ]
)

final_deduplicated_claims = load_jsonl(
    final_registry_required_paths[
        "deduplicated_claims"
    ]
)

final_evidence_links = load_jsonl(
    final_registry_required_paths[
        "evidence_links"
    ]
)

final_claim_relations = load_jsonl(
    final_registry_required_paths[
        "claim_relations"
    ]
)

final_claim_id_mapping = load_jsonl(
    final_registry_required_paths[
        "claim_id_mapping"
    ]
)

final_knowledge_graph = load_json(
    final_registry_required_paths[
        "knowledge_graph"
    ]
)


# =========================================================
# Validate frozen Contract structure
# =========================================================

required_contract_top_level_keys = {
    "contract_version",
    "purpose",
    "source_of_truth",
    "registry_population_policy",
    "artifact_namespace",
    "artifact_family",
    "identity_policy",
    "raw_claim_schema",
    "evidence_link_schema",
    "dedup_relation_types",
    "knowledge_relation_types",
    "relation_policy",
    "consumer_contract",
    "source_corpus_policy",
    "build_policy",
    "current_state",
}


missing_contract_keys = sorted(
    required_contract_top_level_keys
    - set(
        final_contract
    )
)


assert not missing_contract_keys, (
    "Final Registry Contract is missing keys: "
    + ", ".join(
        missing_contract_keys
    )
)


assert (
    final_contract[
        "contract_version"
    ]
    == "claim_registry_contract_v1"
)


assert isinstance(
    final_contract[
        "artifact_family"
    ],
    dict,
)


# =========================================================
# Contract Artifact Family validation
#
# IMPORTANT FIX:
# The actual frozen Contract uses:
#
#   final_contract["artifact_family"]
#
# NOT:
#
#   final_contract["artifacts"]
# =========================================================

final_contract_artifact_family = (
    final_contract[
        "artifact_family"
    ]
)


required_contract_artifact_roles = {
    "raw_claims",
    "validated_claims",
    "deduplicated_claims",
    "evidence_links",
    "claim_relations",
    "knowledge_graph",
    "registry_report",
}


assert (
    required_contract_artifact_roles
    <= set(
        final_contract_artifact_family
    )
), (
    "Frozen Registry Contract is missing "
    "required artifact-family roles."
)


final_contract_artifact_names = set(
    final_contract_artifact_family.values()
)


required_final_contract_artifacts = {
    "claims_raw_v1.jsonl",
    "claims_validated_v1.jsonl",
    "claims_deduplicated_v1.jsonl",
    "claim_evidence_links_v1.jsonl",
    "claim_relations_v1.jsonl",
    "knowledge_graph_v1.json",
    "claim_registry_report_v1.json",
}


assert (
    required_final_contract_artifacts
    <= final_contract_artifact_names
), (
    "Final Registry artifact names do not "
    "match the frozen artifact_family contract."
)


# Validate exact expected role -> filename mapping.
expected_contract_artifact_family = {
    "raw_claims": (
        "claims_raw_v1.jsonl"
    ),

    "validated_claims": (
        "claims_validated_v1.jsonl"
    ),

    "deduplicated_claims": (
        "claims_deduplicated_v1.jsonl"
    ),

    "evidence_links": (
        "claim_evidence_links_v1.jsonl"
    ),

    "claim_relations": (
        "claim_relations_v1.jsonl"
    ),

    "knowledge_graph": (
        "knowledge_graph_v1.json"
    ),

    "registry_report": (
        "claim_registry_report_v1.json"
    ),
}


for (
    artifact_role,
    expected_filename,
) in (
    expected_contract_artifact_family.items()
):

    assert (
        final_contract_artifact_family[
            artifact_role
        ]
        == expected_filename
    ), (
        "Frozen Contract artifact mismatch: "
        f"{artifact_role}"
    )


# =========================================================
# Contract Policy validation
# =========================================================

assert (
    final_contract[
        "identity_policy"
    ][
        "early_canonicalization_allowed"
    ]
    is False
)


assert (
    final_contract[
        "relation_policy"
    ][
        "automatic_semantic_merge_allowed"
    ]
    is False
)


assert (
    final_contract[
        "relation_policy"
    ][
        "automatic_subsumption_merge_allowed"
    ]
    is False
)


assert (
    final_contract[
        "relation_policy"
    ][
        "source_identity_must_be_preserved"
    ]
    is True
)


assert (
    final_contract[
        "relation_policy"
    ][
        "evidence_multiplicity_must_be_preserved"
    ]
    is True
)


assert (
    final_contract[
        "relation_policy"
    ][
        "relation_requires_validated_stage"
    ]
    is True
)


assert (
    final_contract[
        "source_corpus_policy"
    ][
        "claim_registry_replaces_source_corpus"
    ]
    is False
)


assert (
    final_contract[
        "source_corpus_policy"
    ][
        "original_source_remains_final_evidence"
    ]
    is True
)


assert (
    final_contract[
        "source_corpus_policy"
    ][
        "retrieval_chunks_remain_separate"
    ]
    is True
)


assert (
    set(
        final_contract[
            "dedup_relation_types"
        ]
    )
    == {
        "EXACT_DUPLICATE",
        "SEMANTIC_DUPLICATE",
        "SUBSUMES",
        "PARTIAL_OVERLAP",
    }
)


# =========================================================
# SHA-256 manifest
# =========================================================

final_registry_sha256 = {
    artifact_name: file_sha256(
        artifact_path
    )
    for (
        artifact_name,
        artifact_path,
    ) in (
        final_registry_required_paths.items()
    )
}


assert (
    final_registry_sha256[
        "deduplicated_claims"
    ]
    == deduplicated_claims_sha256
)


assert (
    final_registry_sha256[
        "claim_id_mapping"
    ]
    == claim_id_mapping_sha256
)


assert (
    final_registry_sha256[
        "claim_relations"
    ]
    == final_claim_relations_sha256
)


assert (
    final_registry_sha256[
        "knowledge_graph"
    ]
    == knowledge_graph_sha256
)


# =========================================================
# Core population validation
# =========================================================

raw_claim_count = len(
    final_raw_claims
)

validated_claim_count = len(
    final_validated_claims
)

canonical_claim_count = len(
    final_deduplicated_claims
)

evidence_link_count = len(
    final_evidence_links
)

claim_relation_count = len(
    final_claim_relations
)

claim_id_mapping_count = len(
    final_claim_id_mapping
)


assert (
    raw_claim_count
    == 6237
)


assert (
    validated_claim_count
    == 4039
)


assert (
    canonical_claim_count
    == 4006
)


assert (
    evidence_link_count
    == 6396
)


assert (
    claim_relation_count
    == 2516
)


assert (
    claim_id_mapping_count
    == validated_claim_count
)


# =========================================================
# Raw Claim identity validation
# =========================================================

final_raw_claim_ids = [
    item[
        "raw_claim_id"
    ]
    for item in (
        final_raw_claims
    )
]


assert (
    len(
        final_raw_claim_ids
    )
    == len(
        set(
            final_raw_claim_ids
        )
    )
)


final_raw_claim_id_set = set(
    final_raw_claim_ids
)


# =========================================================
# Validated Claim identity validation
# =========================================================

final_validated_raw_claim_ids = [
    item[
        "raw_claim_id"
    ]
    for item in (
        final_validated_claims
    )
]


assert (
    len(
        final_validated_raw_claim_ids
    )
    == len(
        set(
            final_validated_raw_claim_ids
        )
    )
)


final_validated_raw_claim_id_set = set(
    final_validated_raw_claim_ids
)


assert (
    final_validated_raw_claim_id_set
    <= final_raw_claim_id_set
)


# =========================================================
# Canonical Claim identity validation
# =========================================================

final_claim_ids = [
    item[
        "claim_id"
    ]
    for item in (
        final_deduplicated_claims
    )
]


assert (
    len(
        final_claim_ids
    )
    == len(
        set(
            final_claim_ids
        )
    )
)


final_claim_id_set = set(
    final_claim_ids
)


final_canonical_texts = [
    item[
        "claim_text_fa"
    ]
    for item in (
        final_deduplicated_claims
    )
]


assert (
    len(
        final_canonical_texts
    )
    == len(
        set(
            final_canonical_texts
        )
    )
)


# =========================================================
# Verify Claim IDs are reproducible from canonical text
# =========================================================

for claim in (
    final_deduplicated_claims
):

    expected_claim_text_sha256 = (
        hashlib.sha256(
            claim[
                "claim_text_fa"
            ].encode(
                "utf-8"
            )
        ).hexdigest()
    )


    expected_claim_id = (
        "claim:"
        + expected_claim_text_sha256
    )


    assert (
        claim[
            "claim_text_sha256"
        ]
        == expected_claim_text_sha256
    )


    assert (
        claim[
            "claim_id"
        ]
        == expected_claim_id
    )


# =========================================================
# Claim-ID mapping completeness
# =========================================================

mapping_raw_claim_ids = [
    item[
        "raw_claim_id"
    ]
    for item in (
        final_claim_id_mapping
    )
]


assert (
    len(
        mapping_raw_claim_ids
    )
    == len(
        set(
            mapping_raw_claim_ids
        )
    )
)


mapping_raw_claim_id_set = set(
    mapping_raw_claim_ids
)


mapping_claim_id_set = {
    item[
        "claim_id"
    ]
    for item in (
        final_claim_id_mapping
    )
}


assert (
    mapping_raw_claim_id_set
    == final_validated_raw_claim_id_set
)


assert (
    mapping_claim_id_set
    == final_claim_id_set
)


for mapping_record in (
    final_claim_id_mapping
):

    assert (
        mapping_record[
            "claim_id"
        ]
        in final_claim_id_set
    )


# =========================================================
# Evidence validation
# =========================================================

final_evidence_ids = [
    item[
        "evidence_id"
    ]
    for item in (
        final_evidence_links
    )
]


assert (
    len(
        final_evidence_ids
    )
    == len(
        set(
            final_evidence_ids
        )
    )
)


all_evidence_id_set = set(
    final_evidence_ids
)


validated_registry_evidence_ids = {
    evidence_id
    for claim in (
        final_deduplicated_claims
    )
    for evidence_id in (
        claim[
            "evidence_ids"
        ]
    )
}


canonical_evidence_occurrences = sum(
    claim[
        "evidence_count"
    ]
    for claim in (
        final_deduplicated_claims
    )
)


assert (
    len(
        validated_registry_evidence_ids
    )
    == 4125
)


assert (
    canonical_evidence_occurrences
    == 4125
)


assert (
    validated_registry_evidence_ids
    <= all_evidence_id_set
)


# Every canonical Claim must have at least one Evidence.
assert all(
    claim[
        "evidence_count"
    ]
    >= 1
    for claim in (
        final_deduplicated_claims
    )
)


# =========================================================
# Relation validation
# =========================================================

final_relation_ids = [
    item[
        "relation_id"
    ]
    for item in (
        final_claim_relations
    )
]


assert (
    len(
        final_relation_ids
    )
    == len(
        set(
            final_relation_ids
        )
    )
)


final_relation_type_counts = Counter(
    item[
        "relation_type"
    ]
    for item in (
        final_claim_relations
    )
)


expected_final_relation_type_counts = Counter({
    "EXACT_DUPLICATE": 101,
    "PARTIAL_OVERLAP": 1638,
    "SEMANTIC_DUPLICATE": 233,
    "SUBSUMES": 544,
})


assert (
    final_relation_type_counts
    == expected_final_relation_type_counts
)


graph_edge_eligible_relation_count = sum(
    1
    for item in (
        final_claim_relations
    )
    if item[
        "graph_edge_eligible"
    ]
)


canonicalization_provenance_count = sum(
    1
    for item in (
        final_claim_relations
    )
    if not item[
        "graph_edge_eligible"
    ]
)


assert (
    graph_edge_eligible_relation_count
    == 2415
)


assert (
    canonicalization_provenance_count
    == 101
)


# =========================================================
# Validate Relation Claim IDs
# =========================================================

for relation in (
    final_claim_relations
):

    assert (
        relation[
            "claim_id_a"
        ]
        in final_claim_id_set
    )


    assert (
        relation[
            "claim_id_b"
        ]
        in final_claim_id_set
    )


    if (
        relation[
            "relation_type"
        ]
        == "EXACT_DUPLICATE"
    ):

        assert (
            relation[
                "claim_id_a"
            ]
            == relation[
                "claim_id_b"
            ]
        )

        assert (
            relation[
                "graph_edge_eligible"
            ]
            is False
        )


    else:

        assert (
            relation[
                "claim_id_a"
            ]
            != relation[
                "claim_id_b"
            ]
        )

        assert (
            relation[
                "graph_edge_eligible"
            ]
            is True
        )


# =========================================================
# SUBSUMES direction validation
# =========================================================

for relation in (
    final_claim_relations
):

    if (
        relation[
            "relation_type"
        ]
        != "SUBSUMES"
    ):
        continue


    assert (
        relation[
            "subsuming_claim_id"
        ]
        in {
            relation[
                "claim_id_a"
            ],
            relation[
                "claim_id_b"
            ],
        }
    )


    assert (
        relation[
            "subsumed_claim_id"
        ]
        in {
            relation[
                "claim_id_a"
            ],
            relation[
                "claim_id_b"
            ],
        }
    )


    assert (
        relation[
            "subsuming_claim_id"
        ]
        != relation[
            "subsumed_claim_id"
        ]
    )


# =========================================================
# No Semantic self-relations
# =========================================================

semantic_self_relation_count = sum(
    1
    for item in (
        final_claim_relations
    )
    if (
        item[
            "relation_type"
        ]
        != "EXACT_DUPLICATE"
        and item[
            "claim_id_a"
        ]
        == item[
            "claim_id_b"
        ]
    )
)


assert (
    semantic_self_relation_count
    == 0
)


# =========================================================
# Knowledge Graph structural validation
# =========================================================

assert (
    final_knowledge_graph[
        "graph_version"
    ]
    == "knowledge_graph_v1"
)


kg_statistics = (
    final_knowledge_graph[
        "statistics"
    ]
)


assert (
    kg_statistics[
        "claim_nodes"
    ]
    == canonical_claim_count
)


assert (
    kg_statistics[
        "evidence_nodes"
    ]
    == 4125
)


assert (
    kg_statistics[
        "source_nodes"
    ]
    == 667
)


assert (
    kg_statistics[
        "claim_relation_edges"
    ]
    == 2415
)


assert (
    kg_statistics[
        "claim_evidence_edges"
    ]
    == 4125
)


assert (
    kg_statistics[
        "evidence_source_edges"
    ]
    == 4125
)


assert (
    kg_statistics[
        "exact_duplicate_provenance_relations"
    ]
    == 101
)


# =========================================================
# Knowledge Graph source SHA-chain
# =========================================================

assert (
    final_knowledge_graph[
        "source_artifacts"
    ][
        "claims"
    ][
        "sha256"
    ]
    == final_registry_sha256[
        "deduplicated_claims"
    ]
)


assert (
    final_knowledge_graph[
        "source_artifacts"
    ][
        "claim_id_mapping"
    ][
        "sha256"
    ]
    == final_registry_sha256[
        "claim_id_mapping"
    ]
)


assert (
    final_knowledge_graph[
        "source_artifacts"
    ][
        "evidence_links"
    ][
        "sha256"
    ]
    == final_registry_sha256[
        "evidence_links"
    ]
)


assert (
    final_knowledge_graph[
        "source_artifacts"
    ][
        "claim_relations"
    ][
        "sha256"
    ]
    == final_registry_sha256[
        "claim_relations"
    ]
)


# =========================================================
# Knowledge Graph policy validation
# =========================================================

assert (
    final_knowledge_graph[
        "relation_scope"
    ][
        "future_knowledge_relations_inferred"
    ]
    is False
)


assert (
    final_knowledge_graph[
        "relation_scope"
    ][
        "generated_knowledge_relation_types"
    ]
    == []
)


assert (
    final_knowledge_graph[
        "policies"
    ][
        "automatic_semantic_merge_applied"
    ]
    is False
)


assert (
    final_knowledge_graph[
        "policies"
    ][
        "automatic_subsumption_merge_applied"
    ]
    is False
)


assert (
    final_knowledge_graph[
        "policies"
    ][
        "original_source_remains_final_evidence"
    ]
    is True
)


assert (
    final_knowledge_graph[
        "policies"
    ][
        "retrieval_chunks_remain_separate"
    ]
    is True
)


# =========================================================
# Knowledge Graph node coverage
# =========================================================

kg_claim_node_ids = {
    item[
        "claim_id"
    ]
    for item in (
        final_knowledge_graph[
            "nodes"
        ][
            "claims"
        ]
    )
}


assert (
    kg_claim_node_ids
    == final_claim_id_set
)


kg_evidence_node_ids = {
    item[
        "evidence_id"
    ]
    for item in (
        final_knowledge_graph[
            "nodes"
        ][
            "evidence"
        ]
    )
}


assert (
    kg_evidence_node_ids
    == validated_registry_evidence_ids
)


# =========================================================
# Acceptance checks
# =========================================================

final_registry_acceptance_checks = {
    "contract_artifact_family_valid": (
        expected_contract_artifact_family.items()
        <= final_contract_artifact_family.items()
    ),

    "raw_claims_present": (
        raw_claim_count > 0
    ),

    "validated_claims_present": (
        validated_claim_count > 0
    ),

    "canonical_claims_present": (
        canonical_claim_count > 0
    ),

    "all_validated_claims_mapped": (
        mapping_raw_claim_id_set
        == final_validated_raw_claim_id_set
    ),

    "claim_ids_unique": (
        len(
            final_claim_ids
        )
        == len(
            set(
                final_claim_ids
            )
        )
    ),

    "claim_ids_reproducible": all(
        claim[
            "claim_id"
        ]
        == (
            "claim:"
            + hashlib.sha256(
                claim[
                    "claim_text_fa"
                ].encode(
                    "utf-8"
                )
            ).hexdigest()
        )
        for claim in (
            final_deduplicated_claims
        )
    ),

    "canonical_claim_texts_unique": (
        len(
            final_canonical_texts
        )
        == len(
            set(
                final_canonical_texts
            )
        )
    ),

    "validated_evidence_preserved": (
        canonical_evidence_occurrences
        == 4125
    ),

    "all_canonical_claims_have_evidence": all(
        claim[
            "evidence_count"
        ]
        >= 1
        for claim in (
            final_deduplicated_claims
        )
    ),

    "relations_projected": (
        claim_relation_count
        == 2516
    ),

    "graph_edge_relation_count_valid": (
        graph_edge_eligible_relation_count
        == 2415
    ),

    "exact_provenance_relation_count_valid": (
        canonicalization_provenance_count
        == 101
    ),

    "semantic_self_relations_zero": (
        semantic_self_relation_count
        == 0
    ),

    "knowledge_graph_claim_coverage": (
        kg_statistics[
            "claim_nodes"
        ]
        == canonical_claim_count
    ),

    "knowledge_graph_evidence_coverage": (
        kg_statistics[
            "evidence_nodes"
        ]
        == 4125
    ),

    "knowledge_graph_relation_coverage": (
        kg_statistics[
            "claim_relation_edges"
        ]
        == graph_edge_eligible_relation_count
    ),

    "automatic_semantic_merge_disabled": (
        final_knowledge_graph[
            "policies"
        ][
            "automatic_semantic_merge_applied"
        ]
        is False
    ),

    "automatic_subsumption_merge_disabled": (
        final_knowledge_graph[
            "policies"
        ][
            "automatic_subsumption_merge_applied"
        ]
        is False
    ),

    "original_source_remains_evidence": (
        final_knowledge_graph[
            "policies"
        ][
            "original_source_remains_final_evidence"
        ]
        is True
    ),

    "retrieval_chunks_remain_separate": (
        final_knowledge_graph[
            "policies"
        ][
            "retrieval_chunks_remain_separate"
        ]
        is True
    ),

    "future_knowledge_relations_not_fabricated": (
        final_knowledge_graph[
            "relation_scope"
        ][
            "future_knowledge_relations_inferred"
        ]
        is False
    ),
}


failed_final_acceptance_checks = [
    check_name
    for (
        check_name,
        passed,
    ) in (
        final_registry_acceptance_checks.items()
    )
    if not passed
]


assert not failed_final_acceptance_checks, (
    "Final Registry acceptance failed:\n- "
    + "\n- ".join(
        failed_final_acceptance_checks
    )
)


# =========================================================
# Final deterministic Registry Report
#
# No run-local telemetry is stored here.
# Therefore successful reruns remain byte-identical.
# =========================================================

final_registry_report = {
    "report_version": (
        FINAL_REGISTRY_REPORT_VERSION
    ),

    "registry_status": (
        FINAL_REGISTRY_STATUS
    ),

    "registry_role": (
        "VALIDATED_DEDUPLICATED_EVIDENCE_"
        "GROUNDED_CLAIM_REGISTRY"
    ),

    "contract": {
        "contract_version": (
            final_contract[
                "contract_version"
            ]
        ),

        "contract_sha256": (
            final_registry_sha256[
                "contract"
            ]
        ),

        "artifact_namespace": (
            final_contract[
                "artifact_namespace"
            ]
        ),

        "artifact_family": (
            final_contract_artifact_family
        ),
    },

    "counts": {
        "raw_claims": (
            raw_claim_count
        ),

        "validated_claims": (
            validated_claim_count
        ),

        "canonical_claims": (
            canonical_claim_count
        ),

        "raw_evidence_links": (
            evidence_link_count
        ),

        "validated_registry_evidence_occurrences": (
            canonical_evidence_occurrences
        ),

        "claim_id_mapping_records": (
            claim_id_mapping_count
        ),

        "claim_relations_total": (
            claim_relation_count
        ),

        "graph_edge_eligible_claim_relations": (
            graph_edge_eligible_relation_count
        ),

        "exact_duplicate_provenance_relations": (
            canonicalization_provenance_count
        ),

        "knowledge_graph_claim_nodes": (
            kg_statistics[
                "claim_nodes"
            ]
        ),

        "knowledge_graph_evidence_nodes": (
            kg_statistics[
                "evidence_nodes"
            ]
        ),

        "knowledge_graph_source_nodes": (
            kg_statistics[
                "source_nodes"
            ]
        ),

        "knowledge_graph_claim_relation_edges": (
            kg_statistics[
                "claim_relation_edges"
            ]
        ),
    },

    "relation_type_counts": dict(
        sorted(
            final_relation_type_counts.items()
        )
    ),

    "policies": {
        "exact_duplicate_canonicalization": (
            True
        ),

        "automatic_semantic_duplicate_merge": (
            False
        ),

        "automatic_subsumption_merge": (
            False
        ),

        "source_identity_preserved": (
            True
        ),

        "evidence_multiplicity_preserved": (
            True
        ),

        "original_source_remains_final_evidence": (
            True
        ),

        "retrieval_chunks_separate_from_registry_truth": (
            True
        ),

        "future_knowledge_relations_inferred": (
            False
        ),
    },

    "acceptance_checks": (
        final_registry_acceptance_checks
    ),

    "artifact_sha256": (
        final_registry_sha256
    ),

    "api_calls_in_final_report_stage": 0,

    "qwen_required_after_this_stage": False,

    "next_stage": (
        "RESTART_KERNEL_RUN_ALL_AND_NOTEBOOK_HEALTH"
    ),
}


# =========================================================
# Serialize Final Registry Report
# =========================================================

final_registry_report_text = (
    json.dumps(
        final_registry_report,
        ensure_ascii=False,
        indent=2,
        sort_keys=True,
    )
    + "\n"
)


final_registry_report_sha256 = (
    hashlib.sha256(
        final_registry_report_text.encode(
            "utf-8"
        )
    ).hexdigest()
)


final_registry_report_write_status = (
    publish_frozen_artifact(
        final_registry_report_path,
        final_registry_report_text,
    )
)


# =========================================================
# Independent disk revalidation
# =========================================================

disk_final_registry_report = load_json(
    final_registry_report_path
)


assert (
    disk_final_registry_report
    == final_registry_report
)


assert (
    file_sha256(
        final_registry_report_path
    )
    == final_registry_report_sha256
)


assert (
    disk_final_registry_report[
        "registry_status"
    ]
    == "PASS"
)


assert all(
    disk_final_registry_report[
        "acceptance_checks"
    ].values()
)


# =========================================================
# Final Artifact Acceptance Gate
# =========================================================

final_claim_registry_status = (
    "PASS"
)


notebook_run_all_readiness = (
    "READY_FOR_RESTART_KERNEL_RUN_ALL"
)


# =========================================================
# Result
# =========================================================

print(
    "=== NB12i FINAL CLAIM REGISTRY ACCEPTANCE ==="
)

print()

print(
    "Contract version:",
    final_contract[
        "contract_version"
    ],
)

print(
    "Contract artifact family:",
    len(
        final_contract_artifact_family
    ),
    "artifacts",
)


print()

print(
    "Raw Claims:",
    raw_claim_count,
)

print(
    "Validated Claims:",
    validated_claim_count,
)

print(
    "Canonical Claims:",
    canonical_claim_count,
)

print(
    "Validated Evidence occurrences:",
    canonical_evidence_occurrences,
)


print()

print(
    "Claim Relations:",
    claim_relation_count,
)

print(
    "Graph-edge eligible Relations:",
    graph_edge_eligible_relation_count,
)

print(
    "Exact Duplicate provenance:",
    canonicalization_provenance_count,
)

print(
    "Relation types:",
    dict(
        sorted(
            final_relation_type_counts.items()
        )
    ),
)


print()

print(
    "Knowledge Graph Claim Nodes:",
    kg_statistics[
        "claim_nodes"
    ],
)

print(
    "Knowledge Graph Evidence Nodes:",
    kg_statistics[
        "evidence_nodes"
    ],
)

print(
    "Knowledge Graph Source Nodes:",
    kg_statistics[
        "source_nodes"
    ],
)

print(
    "Knowledge Graph Claim Relation Edges:",
    kg_statistics[
        "claim_relation_edges"
    ],
)


print()

print(
    "Acceptance checks:",
    sum(
        final_registry_acceptance_checks.values()
    ),
    "/",
    len(
        final_registry_acceptance_checks
    ),
    "PASS",
)


print()

print(
    "Semantic self-relations:",
    semantic_self_relation_count,
)

print(
    "Automatic Semantic Merge:",
    False,
)

print(
    "Automatic Subsumption Merge:",
    False,
)

print(
    "Future Knowledge Relations fabricated:",
    False,
)

print(
    "API calls performed:",
    0,
)

print(
    "Qwen required after this stage:",
    False,
)


print()

print(
    "Final Registry Report:",
    final_registry_report_path.relative_to(
        project_root
    ),
)

print(
    "Final Registry Report SHA-256:",
    final_registry_report_sha256,
)

print(
    "Write status:",
    final_registry_report_write_status,
)


print()

print(
    "Registry Status:",
    final_claim_registry_status,
)

print(
    "Next stage:",
    "RESTART_KERNEL_RUN_ALL_AND_NOTEBOOK_HEALTH",
)

print(
    "Run-All readiness:",
    notebook_run_all_readiness,
)


print()

print(
    "CODE CELL 30 GATE:",
    final_claim_registry_status,
)
```

```python
# Code Cell 31

assert (
    final_claim_registry_status
    == "PASS"
), (
    "Code Cell 30 must PASS before "
    "Final Notebook Health Check."
)

assert (
    notebook_run_all_readiness
    == "READY_FOR_RESTART_KERNEL_RUN_ALL"
)


# =========================================================
# Configuration
# =========================================================

EXPECTED_NOTEBOOK_RELATIVE_PATH = Path(
    "notebooks/phase2/"
    "12i_claim_registry_and_knowledge_graph.ipynb"
)


final_notebook_path = (
    project_root
    / EXPECTED_NOTEBOOK_RELATIVE_PATH
)


assert (
    final_notebook_path.exists()
), (
    "Actual NB12i .ipynb file not found: "
    f"{final_notebook_path}"
)


# =========================================================
# Parse actual Notebook file
# =========================================================

with final_notebook_path.open(
    "r",
    encoding="utf-8",
) as file:

    final_notebook_json = json.load(
        file
    )


final_notebook_cells = (
    final_notebook_json[
        "cells"
    ]
)


final_code_cells = [
    cell
    for cell in final_notebook_cells
    if (
        cell.get(
            "cell_type"
        )
        == "code"
    )
]


final_markdown_cells = [
    cell
    for cell in final_notebook_cells
    if (
        cell.get(
            "cell_type"
        )
        == "markdown"
    )
]


assert (
    len(
        final_code_cells
    )
    == 31
), (
    "Expected exactly 31 Code Cells, found "
    f"{len(final_code_cells)}."
)


# =========================================================
# Source helpers
# =========================================================

def notebook_cell_source(
    cell,
):

    source = cell.get(
        "source",
        "",
    )

    if isinstance(
        source,
        list,
    ):

        source = "".join(
            source
        )

    assert isinstance(
        source,
        str,
    )

    return source


def normalize_code_source(
    source,
):

    source = source.replace(
        "\r\n",
        "\n",
    ).replace(
        "\r",
        "\n",
    )

    return "\n".join(
        line.rstrip()
        for line in (
            source.strip().splitlines()
        )
    )


final_code_sources = [
    notebook_cell_source(
        cell
    )
    for cell in final_code_cells
]


# =========================================================
# Code Cell numbering
# =========================================================

code_cell_numbers = []


for (
    position,
    source,
) in enumerate(
    final_code_sources,
    start=1,
):

    match = re.search(
        r"(?m)^\s*#\s*Code Cell\s+(\d+)\s*$",
        source,
    )

    assert (
        match
        is not None
    ), (
        "Missing '# Code Cell XX' marker "
        f"in executable Cell position {position}."
    )

    code_cell_numbers.append(
        int(
            match.group(
                1
            )
        )
    )


expected_code_cell_numbers = list(
    range(
        1,
        32,
    )
)


assert (
    code_cell_numbers
    == expected_code_cell_numbers
), (
    "Code Cell numbering is not sequential.\n"
    f"Expected: {expected_code_cell_numbers}\n"
    f"Actual:   {code_cell_numbers}"
)


# =========================================================
# Cell 01 central import validation
# =========================================================

bootstrap_source = (
    final_code_sources[
        0
    ]
)


required_bootstrap_import_fragments = [
    "from pathlib import Path",
    "from collections import Counter, defaultdict",
    "ThreadPoolExecutor",
    "as_completed",
    "import hashlib",
    "import json",
    "import os",
    "import re",
    "import subprocess",
    "import threading",
    "import time",
    "import numpy as np",
    "from openai import OpenAI",
]


missing_bootstrap_imports = [
    fragment
    for fragment in (
        required_bootstrap_import_fragments
    )
    if (
        fragment
        not in bootstrap_source
    )
]


assert not missing_bootstrap_imports, (
    "Code Cell 01 is missing centralized imports:\n- "
    + "\n- ".join(
        missing_bootstrap_imports
    )
)


# =========================================================
# Clean Restart Kernel -> Run All validation
#
# Current Kernel history must contain exactly the 31 Code
# Cells in Notebook order. Therefore no manual Cell rerun
# occurred in this Kernel before Health Check.
# =========================================================

current_ipython = get_ipython()


assert (
    current_ipython
    is not None
), (
    "Final Health Check requires Jupyter/IPython."
)


current_input_history = (
    current_ipython.user_ns.get(
        "In",
        [],
    )
)


current_kernel_inputs = [
    item
    for item in (
        current_input_history[
            1:
        ]
    )
    if isinstance(
        item,
        str,
    )
]


assert (
    len(
        current_kernel_inputs
    )
    == 31
), (
    "This is not a clean single Restart Kernel -> Run All.\n"
    f"Expected Kernel executions: 31\n"
    f"Actual Kernel executions: "
    f"{len(current_kernel_inputs)}"
)


normalized_notebook_sources = [
    normalize_code_source(
        source
    )
    for source in (
        final_code_sources
    )
]


normalized_kernel_inputs = [
    normalize_code_source(
        source
    )
    for source in (
        current_kernel_inputs
    )
]


execution_source_mismatches = []


for index, (
    notebook_source,
    executed_source,
) in enumerate(
    zip(
        normalized_notebook_sources,
        normalized_kernel_inputs,
    ),
    start=1,
):

    if (
        notebook_source
        != executed_source
    ):

        execution_source_mismatches.append(
            index
        )


assert not execution_source_mismatches, (
    "Kernel execution history does not match "
    "Notebook Cell order.\n"
    "Mismatched Code Cells: "
    + ", ".join(
        str(
            item
        )
        for item in (
            execution_source_mismatches
        )
    )
)


# =========================================================
# Critical runtime gates
#
# These gates prove that the major pipeline stages reached
# their accepted terminal state during THIS Kernel run.
# =========================================================

critical_runtime_gates = {
    "semantic_relation_pilot_status": (
        semantic_relation_pilot_status
    ),

    "semantic_relation_full_status": (
        semantic_relation_full_status
    ),

    "dedup_relation_consolidation_status": (
        dedup_relation_consolidation_status
    ),

    "exact_canonical_registry_status": (
        exact_canonical_registry_status
    ),

    "claim_relation_projection_status": (
        claim_relation_projection_status
    ),

    "knowledge_graph_build_status": (
        knowledge_graph_build_status
    ),

    "final_claim_registry_status": (
        final_claim_registry_status
    ),
}


failed_runtime_gates = {
    gate_name: gate_value
    for (
        gate_name,
        gate_value,
    ) in (
        critical_runtime_gates.items()
    )
    if (
        gate_value
        != "PASS"
    )
}


assert not failed_runtime_gates, (
    "Critical NB12i runtime gate failure: "
    f"{failed_runtime_gates}"
)


# =========================================================
# Required final artifacts
# =========================================================

health_required_artifacts = {
    "claims_raw_v1.jsonl": (
        registry_artifact_root
        / "claims_raw_v1.jsonl"
    ),

    "claims_validated_v1.jsonl": (
        registry_artifact_root
        / "claims_validated_v1.jsonl"
    ),

    "claims_deduplicated_v1.jsonl": (
        registry_artifact_root
        / "claims_deduplicated_v1.jsonl"
    ),

    "claim_evidence_links_v1.jsonl": (
        registry_artifact_root
        / "claim_evidence_links_v1.jsonl"
    ),

    "claim_relations_v1.jsonl": (
        registry_artifact_root
        / "claim_relations_v1.jsonl"
    ),

    "knowledge_graph_v1.json": (
        registry_artifact_root
        / "knowledge_graph_v1.json"
    ),

    "claim_registry_report_v1.json": (
        registry_artifact_root
        / "claim_registry_report_v1.json"
    ),
}


missing_health_artifacts = [
    artifact_name
    for (
        artifact_name,
        artifact_path,
    ) in (
        health_required_artifacts.items()
    )
    if (
        not artifact_path.exists()
    )
]


assert not missing_health_artifacts, (
    "Final Registry artifacts missing:\n- "
    + "\n- ".join(
        missing_health_artifacts
    )
)


# =========================================================
# Reload Final Registry Report
# =========================================================

health_registry_report = load_json(
    health_required_artifacts[
        "claim_registry_report_v1.json"
    ]
)


assert (
    health_registry_report[
        "registry_status"
    ]
    == "PASS"
)


assert all(
    health_registry_report[
        "acceptance_checks"
    ].values()
)


# =========================================================
# Final population checks
# =========================================================

assert (
    health_registry_report[
        "counts"
    ][
        "raw_claims"
    ]
    == 6237
)


assert (
    health_registry_report[
        "counts"
    ][
        "validated_claims"
    ]
    == 4039
)


assert (
    health_registry_report[
        "counts"
    ][
        "canonical_claims"
    ]
    == 4006
)


assert (
    health_registry_report[
        "counts"
    ][
        "validated_registry_evidence_occurrences"
    ]
    == 4125
)


assert (
    health_registry_report[
        "counts"
    ][
        "claim_relations_total"
    ]
    == 2516
)


assert (
    health_registry_report[
        "counts"
    ][
        "graph_edge_eligible_claim_relations"
    ]
    == 2415
)


assert (
    health_registry_report[
        "counts"
    ][
        "exact_duplicate_provenance_relations"
    ]
    == 101
)


# =========================================================
# Final relation checks
# =========================================================

expected_relation_counts = {
    "EXACT_DUPLICATE": 101,
    "PARTIAL_OVERLAP": 1638,
    "SEMANTIC_DUPLICATE": 233,
    "SUBSUMES": 544,
}


assert (
    health_registry_report[
        "relation_type_counts"
    ]
    == expected_relation_counts
)


# =========================================================
# Policy checks
# =========================================================

health_policies = (
    health_registry_report[
        "policies"
    ]
)


assert (
    health_policies[
        "exact_duplicate_canonicalization"
    ]
    is True
)


assert (
    health_policies[
        "automatic_semantic_duplicate_merge"
    ]
    is False
)


assert (
    health_policies[
        "automatic_subsumption_merge"
    ]
    is False
)


assert (
    health_policies[
        "source_identity_preserved"
    ]
    is True
)


assert (
    health_policies[
        "evidence_multiplicity_preserved"
    ]
    is True
)


assert (
    health_policies[
        "original_source_remains_final_evidence"
    ]
    is True
)


assert (
    health_policies[
        "retrieval_chunks_separate_from_registry_truth"
    ]
    is True
)


assert (
    health_policies[
        "future_knowledge_relations_inferred"
    ]
    is False
)


# =========================================================
# Artifact SHA validation
# =========================================================

health_artifact_sha256 = {
    artifact_name: file_sha256(
        artifact_path
    )
    for (
        artifact_name,
        artifact_path,
    ) in (
        health_required_artifacts.items()
    )
}


assert (
    health_artifact_sha256[
        "claims_deduplicated_v1.jsonl"
    ]
    == deduplicated_claims_sha256
)


assert (
    health_artifact_sha256[
        "claim_relations_v1.jsonl"
    ]
    == final_claim_relations_sha256
)


assert (
    health_artifact_sha256[
        "knowledge_graph_v1.json"
    ]
    == knowledge_graph_sha256
)


assert (
    health_artifact_sha256[
        "claim_registry_report_v1.json"
    ]
    == final_registry_report_sha256
)


# =========================================================
# Independently validate final Knowledge Graph
# =========================================================

health_knowledge_graph = load_json(
    health_required_artifacts[
        "knowledge_graph_v1.json"
    ]
)


health_kg_statistics = (
    health_knowledge_graph[
        "statistics"
    ]
)


assert (
    health_kg_statistics[
        "claim_nodes"
    ]
    == 4006
)


assert (
    health_kg_statistics[
        "evidence_nodes"
    ]
    == 4125
)


assert (
    health_kg_statistics[
        "source_nodes"
    ]
    == 667
)


assert (
    health_kg_statistics[
        "claim_relation_edges"
    ]
    == 2415
)


assert (
    health_kg_statistics[
        "claim_evidence_edges"
    ]
    == 4125
)


assert (
    health_kg_statistics[
        "evidence_source_edges"
    ]
    == 4125
)


assert (
    health_kg_statistics[
        "exact_duplicate_provenance_relations"
    ]
    == 101
)


# =========================================================
# Notebook source-only SHA
# =========================================================

notebook_source_manifest = [
    {
        "cell_type": (
            cell.get(
                "cell_type"
            )
        ),

        "source": (
            notebook_cell_source(
                cell
            )
        ),
    }
    for cell in (
        final_notebook_cells
    )
]


notebook_source_manifest_text = (
    json.dumps(
        notebook_source_manifest,
        ensure_ascii=False,
        sort_keys=True,
    )
)


final_notebook_source_sha256 = (
    hashlib.sha256(
        notebook_source_manifest_text.encode(
            "utf-8"
        )
    ).hexdigest()
)


# =========================================================
# Final Health checks
# =========================================================

final_notebook_health_checks = {
    "actual_ipynb_parsed": True,

    "code_cell_count_31": (
        len(
            final_code_cells
        )
        == 31
    ),

    "code_cell_numbering_sequential": (
        code_cell_numbers
        == expected_code_cell_numbers
    ),

    "cell01_imports_centralized": (
        not missing_bootstrap_imports
    ),

    "clean_kernel_single_run_all": (
        len(
            current_kernel_inputs
        )
        == 31
    ),

    "execution_order_matches_notebook": (
        not execution_source_mismatches
    ),

    "critical_pipeline_gates_pass": (
        not failed_runtime_gates
    ),

    "final_registry_acceptance_pass": (
        health_registry_report[
            "registry_status"
        ]
        == "PASS"
    ),

    "all_registry_acceptance_checks_pass": (
        all(
            health_registry_report[
                "acceptance_checks"
            ].values()
        )
    ),

    "required_final_artifacts_present": (
        not missing_health_artifacts
    ),

    "final_population_counts_valid": True,

    "final_relation_counts_valid": True,

    "final_policy_contract_valid": True,

    "final_artifact_hashes_valid": True,

    "knowledge_graph_structure_valid": True,

    "qwen_required_after_nb12i": (
        health_registry_report[
            "qwen_required_after_this_stage"
        ]
        is False
    ),
}


failed_notebook_health_checks = [
    check_name
    for (
        check_name,
        passed,
    ) in (
        final_notebook_health_checks.items()
    )
    if not passed
]


assert not failed_notebook_health_checks, (
    "Final Notebook Health Check failed:\n- "
    + "\n- ".join(
        failed_notebook_health_checks
    )
)


# =========================================================
# Final Objective Status
# =========================================================

final_notebook_health_status = (
    "PASS"
)

final_nb12i_status = (
    "PASS"
)

final_nb12i_goal_status = (
    "GOAL_ACHIEVED"
)


# =========================================================
# Result
# =========================================================

print(
    "=== NB12i FINAL NOTEBOOK HEALTH ==="
)

print()

print(
    "Notebook:",
    final_notebook_path.relative_to(
        project_root
    ),
)

print(
    "Notebook source SHA-256:",
    final_notebook_source_sha256,
)


print()

print(
    "Markdown Cells:",
    len(
        final_markdown_cells
    ),
)

print(
    "Code Cells:",
    len(
        final_code_cells
    ),
)

print(
    "Code Cell numbering:",
    "01 -> 31",
)


print()

print(
    "Kernel executions:",
    len(
        current_kernel_inputs
    ),
)

print(
    "Execution order matches Notebook:",
    (
        len(
            execution_source_mismatches
        )
        == 0
    ),
)

print(
    "Critical pipeline gates:",
    len(
        critical_runtime_gates
    ),
    "/",
    len(
        critical_runtime_gates
    ),
    "PASS",
)


print()

print(
    "Final Registry acceptance checks:",
    sum(
        health_registry_report[
            "acceptance_checks"
        ].values()
    ),
    "/",
    len(
        health_registry_report[
            "acceptance_checks"
        ]
    ),
    "PASS",
)


print()

print(
    "Raw Claims:",
    health_registry_report[
        "counts"
    ][
        "raw_claims"
    ],
)

print(
    "Validated Claims:",
    health_registry_report[
        "counts"
    ][
        "validated_claims"
    ],
)

print(
    "Canonical Claims:",
    health_registry_report[
        "counts"
    ][
        "canonical_claims"
    ],
)

print(
    "Validated Evidence occurrences:",
    health_registry_report[
        "counts"
    ][
        "validated_registry_evidence_occurrences"
    ],
)

print(
    "Claim Relations:",
    health_registry_report[
        "counts"
    ][
        "claim_relations_total"
    ],
)

print(
    "Graph-edge eligible Relations:",
    health_registry_report[
        "counts"
    ][
        "graph_edge_eligible_claim_relations"
    ],
)


print()

print(
    "Notebook Health checks:",
    sum(
        final_notebook_health_checks.values()
    ),
    "/",
    len(
        final_notebook_health_checks
    ),
    "PASS",
)


print()

print(
    "Qwen required after NB12i:",
    False,
)

print(
    "FINAL NOTEBOOK HEALTH STATUS:",
    final_notebook_health_status,
)

print(
    "FINAL NB12i STATUS:",
    final_nb12i_status,
)

print(
    "FINAL NB12i GOAL:",
    final_nb12i_goal_status,
)


print()

print(
    "CODE CELL 31 GATE:",
    final_notebook_health_status,
)
```

<h3>هدف Notebook</h3>

<p>
هدف این Notebook ساخت یک Claim Registry گسترده، Evidence-Grounded،
قابل ردیابی و مستقل از Retrieval Chunkها بر روی Corpus پذیرفته‌شده Phase 2 بود؛
به‌گونه‌ای که واحدهای دانشی مستقل کتاب و سایر منابع از متن اصلی استخراج شوند،
اعتبار معنایی و Grounding آن‌ها کنترل شود، روابط Duplicate و Overlap میان آن‌ها
ثبت شود و یک لایه دانشی پایدار برای مصرف مشترک Golden Dataset v2، Book Pipeline
و GraphRAG ساخته شود.
</p>

<p>
در این معماری، Claim Registry جای Source Corpus را نمی‌گیرد و Retrieval Chunk
نیز واحد حقیقت Registry محسوب نمی‌شود. منبع اصلی و Evidence دقیق همچنان
مرجع نهایی هر Claim باقی می‌مانند.
</p>

<h3>نتیجه استخراج و Validation</h3>

<ul>
<li>Corpus پذیرفته‌شده Phase 2 شامل ۱۹۴۵۶ Processing Record بود.</li>
<li>۱۵۱۵۸ Record برای Claim Discovery واجد شرایط بودند.</li>
<li>۲۱۲۹ Structural Extraction Unit ساخته شد.</li>
<li>۲۲۸۷ Prompt Window برای استخراج کنترل‌شده ایجاد شد.</li>
<li>۶۲۳۷ Raw Claim استخراج شد.</li>
<li>پس از Semantic Validation، تعداد ۴۰۳۹ Claim در وضعیت VALIDATED قرار گرفت.</li>
<li>هیچ Claim بدون Evidence وارد Registry نهایی نشد.</li>
</ul>

<h3>نتیجه Deduplication و هویت نهایی Claimها</h3>

<ul>
<li>۴۰۳۹ Claim معتبر به ۴۰۰۶ Exact-Unique Canonical Claim تبدیل شد.</li>
<li>۱۴ Exact Duplicate Group شناسایی شد.</li>
<li>۳۳ Raw Claim فقط از طریق Exact Dedup در Canonical Claimهای مشترک ادغام شدند.</li>
<li>برای هر ۴۰۰۶ Canonical Claim یک <code>claim_id</code> پایدار ساخته شد.</li>
<li>هیچ Semantic Duplicate به‌صورت خودکار Merge نشد.</li>
<li>هیچ SUBSUMES Relation به‌صورت خودکار Merge نشد.</li>
</ul>

<h3>نتیجه Evidence</h3>

<ul>
<li>تمام Evidenceهای Claimهای معتبر حفظ شدند.</li>
<li>۴۱۲۵ Evidence occurrence در Registry نهایی وجود دارد.</li>
<li>Evidence multiplicity و Source identity در Canonicalization از بین نرفت.</li>
<li>Original Source همچنان Final Evidence محسوب می‌شود.</li>
</ul>

<h3>نتیجه Relation Registry</h3>

<p>
در مجموع ۲۵۱۶ Relation نهایی ثبت شد:
</p>

<ul>
<li>۱۰۱ <code>EXACT_DUPLICATE</code> Relation برای Canonicalization Provenance</li>
<li>۲۳۳ <code>SEMANTIC_DUPLICATE</code></li>
<li>۵۴۴ <code>SUBSUMES</code></li>
<li>۱۶۳۸ <code>PARTIAL_OVERLAP</code></li>
</ul>

<p>
۱۰۱ Exact Duplicate Relation پس از Canonicalization به یک Claim ID واحد اشاره
می‌کنند و به‌عنوان provenance حفظ شده‌اند؛ بنابراین Knowledge Graph self-edge
ایجاد نمی‌کنند. ۲۴۱۵ Relation باقی‌مانده Edge واقعی بین Canonical Claimهای مستقل هستند.
</p>

<h3>نتیجه Knowledge Graph</h3>

<ul>
<li>۴۰۰۶ Claim Node</li>
<li>۴۱۲۵ Evidence Node</li>
<li>۶۶۷ Source Node</li>
<li>۲۴۱۵ Claim-to-Claim Relation Edge</li>
<li>۴۱۲۵ Claim-to-Evidence Edge</li>
<li>۴۱۲۵ Evidence-to-Source Edge</li>
<li>صفر Semantic Self-Edge</li>
<li>صفر Canonical Claim بدون Evidence</li>
</ul>

<p>
Relationهای آینده مانند
<code>RELATED_TO</code>،
<code>SUPPORTS</code>،
<code>CONTRADICTS</code> و
<code>PREREQUISITE_OF</code>
در این Notebook جعل یا حدس زده نشدند، زیرا برای آن‌ها Adjudication مستقل انجام نشده است.
</p>
