**نمونهٔ مشترک:** حداکثر ۱۲ سؤال Development؛ تا چهار سؤال کتاب، چهار صوت و چهار وب؛ انتخاب قطعی بدون نگاه به نتیجه. نمونه‌گیر ابتدا از مؤلفه‌های متفاوت انتخاب می‌کند و اگر یک نوع منبع مؤلفهٔ کمتری داشته باشد، از همان مؤلفه‌های واقعی سؤال‌های بیشتر می‌گیرد. مؤلفهٔ مستقل جعل نمی‌شود؛ تعداد واقعی و تکرارها در گزارش می‌آید. اگر تعداد سؤال یکی از سه نوع کمتر از چهار باشد، تعداد هر سه نوع به‌طور متوازن کاهش می‌یابد. این یک پایلوت کوچک برای حذف مسیر نامناسب و سنجش کارایی است، نه آزمون نهایی یا تضمین برندهٔ کل Development. نمونه را پس از دیدن نتیجه عوض نکن.

**اجرا محلی است:** API، هزینهٔ سرویس، دانلود و Test تازه صفر؛ ولی برخلاف NB22D، **inference تازهٔ CPU** روی جفت‌های فاقد امتیاز انجام می‌شود. حداکثر ۱۲۰۰۰ جفت فاقد کش و یک ساعت زمان کل، با ذخیره پس از هر batch چهارتایی؛ عبور از حد، مسیر را Held می‌کند. اگر مدل/نسخه/کش لازم موجود نباشد، دلیل دقیق چاپ و فقط مسیر مدل مسدود می‌شود؛ مرجع مستقل همچنان قابل گزارش است. زمان واقعی روی دستگاه کاربر معلوم می‌شود.

```python
# NB22E-C002 — Shared settings, exact inputs, engine and frozen pilot.
PINS = {'C056': {'path': 'reports/phase2/release/NB22.C056.retrieval_verification/C056.independent_retrieval_metric_verification.ff6d6e136043929e0156e43f6ea75e65bd64f40ab7bcdafef90a4ba40901de22.json', 'file_bytes_sha256': 'ff6d6e136043929e0156e43f6ea75e65bd64f40ab7bcdafef90a4ba40901de22', 'bytes': 1180779, 'write_status': 'CREATED', 'reload_validation': 'PASS'}, 'C_final': {'path': 'reports/phase2/quality_improvement/NB22C.v1/A06.final_report.1797d76fe08136a50ffa846da39858f29838a9bc8a3dc3d1cad120789ccc53ea.json', 'file_bytes_sha256': '1797d76fe08136a50ffa846da39858f29838a9bc8a3dc3d1cad120789ccc53ea', 'bytes': 31539, 'reload_validation': 'PASS'}, 'C_rankings': {'path': 'reports/phase2/quality_improvement/NB22C.v1/A05.retained_development_rankings.5bc216aae839bc80cc2f184ca3a3cd3e4ac33657655c3bffb4fac410e96e9190.jsonl', 'file_bytes_sha256': '5bc216aae839bc80cc2f184ca3a3cd3e4ac33657655c3bffb4fac410e96e9190', 'bytes': 19285045, 'reload_validation': 'PASS'}, 'C_diagnosis': {'path': 'reports/phase2/quality_improvement/NB22C.v1/A02.development_diagnosis.2b86a0709a76b3e5673383d6f95d973991890246cb129faf7fe995e8f28dcdf5.jsonl', 'file_bytes_sha256': '2b86a0709a76b3e5673383d6f95d973991890246cb129faf7fe995e8f28dcdf5', 'bytes': 1610871, 'reload_validation': 'PASS'}, 'D_final': {'path': 'reports/phase2/quality_improvement/NB22D.v1/A05.final_report.4b1626f9669aa7bc8ffa0ada5ddffe8d7350aa8dcf5db9e956bd8c67f86c82de.json', 'file_bytes_sha256': '4b1626f9669aa7bc8ffa0ada5ddffe8d7350aa8dcf5db9e956bd8c67f86c82de', 'bytes': 11977, 'reload_validation': 'PASS'}, 'B_cache_inventory': {'bytes': 760689, 'file_bytes_sha256': '3ebb5c79d71e0ee21ffab7e99a1f32580547a1d1f7a7b0a8a7cb693a294c0e93', 'path': 'reports/phase2/quality_improvement/NB22B.v1/A05.local_score_cache_inventory.3ebb5c79d71e0ee21ffab7e99a1f32580547a1d1f7a7b0a8a7cb693a294c0e93.json', 'reload_validation': 'PASS'}}
MODEL = {'model_id': 'cross-encoder/mmarco-mMiniLMv2-L12-H384-v1', 'revision': '1427fd652930e4ba29e8149678df786c240d8825', 'model_sha256': '6c2513767fb63d008a4377bef7a7a3555433d9436342bb53e35a3a72ffc52d4b', 'onnxruntime_version': '1.16.3', 'batch_size': 4, 'maximum_sequence_length': 512, 'implementation_ast_sha256': 'd6713ab6ba672379e122d0712b2a6f4a7a9dd40eedfb437c5262366cd766f4c5'}
ROUTES = [{'name': 'baseline_D', 'mode': 'baseline'}, {'name': 'onnx200', 'mode': 'rerank', 'depth': 200}, {'name': 'onnx400', 'mode': 'rerank', 'depth': 400}, {'name': 'onnx800', 'mode': 'rerank', 'depth': 800}, {'name': 'onnx400_fusion', 'mode': 'fusion', 'depth': 400}]

# Shared setup: explicit offline CPU pilot, no model download or paid API.
CONFIG={'version':'NB22E_LAB_V2','seed':'NB22E_v2_SOURCE_PILOT',
 'pilot_per_source_type':4,'maximum_new_pairs_per_run':12000,'maximum_local_seconds_per_run':3600,
 'thresholds':{'recall_at_10':.85,'ndcg_at_10':.75},'rrf_k':10,'reranker_weight':.75}
_start=Path(os.environ.get('RAG_FINANCE_ROOT',str(Path.cwd()))).expanduser().resolve()
ROOT=next((p for p in [_start,*_start.parents] if (p/'notebooks/phase2').is_dir() and
 (p/'data/Road_Maps/RAG_finance_phase2_modular_notebooks_execution_plan.md').is_file()),None)
STATE={'actions':{},'artifacts':[],'memory':{},'errors':[],'inputs':{},'api_calls':0,'model_pairs':0,'network_calls':0}
LAB={'ready':False,'results':{},'issues':[],'started':None,'new_pairs':0,'predict':None,'score_memory':{},'new_inference_seconds':0.0,'parity_pairs':0,'last_progress':0.0,'model_ready':False,'model_issues':[]}
NOTEBOOK_CODE_HASH=None;SAVED_CODE_VALID=False
if ROOT is not None:
 p=ROOT/'notebooks/phase2/22e_retrieval_experiment_lab_v2.ipynb'
 if p.is_file():
  n=json.loads(p.read_bytes());codes=[''.join(c['source']) if isinstance(c['source'],list) else c['source'] for c in n['cells'] if c['cell_type']=='code']
  NOTEBOOK_CODE_HASH=hashlib.sha256(json.dumps(codes,ensure_ascii=False,sort_keys=True,separators=(',',':'),allow_nan=False).encode()).hexdigest()
  SAVED_CODE_VALID=NOTEBOOK_CODE_HASH==n.get('metadata',{}).get('nb22e',{}).get('code_source_sha256')

# NB22E — shared mechanics — Shared mechanics; no historical notebook is executed.
class InputGap(RuntimeError):
    pass

def require(ok, reason):
    if not ok:
        raise InputGap(reason)

def canonical(x):
    return json.dumps(x, ensure_ascii=False, sort_keys=True,
                      separators=(',', ':'), allow_nan=False).encode('utf-8')

def digest(raw):
    return hashlib.sha256(raw).hexdigest()

def decode(raw):
    def pairs(items):
        out = {}
        for k, v in items:
            if k in out:
                raise InputGap('DUPLICATE_JSON_KEY:' + k)
            out[k] = v
        return out
    def reject(x):
        raise InputGap('NONFINITE_JSON:' + x)
    return json.loads(raw, object_pairs_hook=pairs, parse_constant=reject)

def path_for(relative):
    require(ROOT is not None, 'PROJECT_ROOT_REQUIRED:run_inside_repository_or_set_RAG_FINANCE_ROOT')
    p = Path(relative)
    require(not p.is_absolute() and '..' not in p.parts, 'PROJECT_RELATIVE_PATH_REQUIRED')
    target = (ROOT / p).resolve()
    require(target.is_relative_to(ROOT), 'PATH_OUTSIDE_PROJECT')
    return target

def binding_hash(binding):
    values = {binding[k] for k in ('file_bytes_sha256', 'sha256') if k in binding}
    require(len(values) == 1, 'UNAMBIGUOUS_BYTE_HASH_REQUIRED')
    h = next(iter(values))
    require(isinstance(h, str) and re.fullmatch('[a-f0-9]{64}', h), 'INVALID_HASH')
    return h

def checked_bytes(binding):
    p = path_for(binding['path'])
    require(p.is_file(), 'REQUIRED_FILE_NOT_AVAILABLE:' + binding['path'])
    before = p.stat()
    raw = p.read_bytes()
    after = p.stat()
    require((before.st_size, before.st_mtime_ns, before.st_ino) ==
            (after.st_size, after.st_mtime_ns, after.st_ino), 'SOURCE_CHANGED_DURING_READ')
    require(digest(raw) == binding_hash(binding), 'INPUT_BYTE_HASH_MISMATCH:' + binding['path'])
    STATE['inputs'][binding['path']] = dict(binding)
    return raw

def read_json(binding):
    return decode(checked_bytes(binding))

def read_rows(binding):
    raw = checked_bytes(binding)
    require(all(line.strip() for line in raw.splitlines()), 'BLANK_JSONL_ROW')
    return [decode(line) for line in raw.splitlines()]

def keyed(rows, field):
    out = {}
    for row in rows:
        require(isinstance(row, dict) and isinstance(row.get(field), str) and
                row[field] not in out, 'MISSING_OR_DUPLICATE_ID:' + field)
        out[row[field]] = row
    return out

def save(stem, value, rows=False):
    folder = path_for('reports/phase2/quality_improvement/NB22E.lab.v2')
    folder.mkdir(parents=True, exist_ok=True)
    raw = b''.join(canonical(r) + b'\n' for r in value) if rows else canonical(value)
    suffix = 'jsonl' if rows else 'json'
    p = folder / (stem + '.' + digest(raw) + '.' + suffix)
    if not p.exists():
        fd, name = tempfile.mkstemp(dir=folder, prefix='.nb22e-')
        try:
            with os.fdopen(fd, 'wb') as f:
                f.write(raw); f.flush(); os.fsync(f.fileno())
            try:
                os.link(name, p)
            except FileExistsError:
                pass
        finally:
            Path(name).unlink(missing_ok=True)
    actual = p.read_bytes()
    require(actual == raw, 'OUTPUT_COLLISION_OR_RELOAD_FAILURE')
    if rows:
        require([decode(x) for x in actual.splitlines()] == value, 'OUTPUT_ROW_RELOAD_FAILURE')
    else:
        require(decode(actual) == value, 'OUTPUT_RELOAD_FAILURE')
    receipt = {'path': p.relative_to(ROOT).as_posix(), 'file_bytes_sha256': digest(raw),
               'bytes': len(raw), 'reload_validation': 'PASS'}
    STATE['artifacts'].append(receipt)
    return receipt

def metrics(order, gold):
    require(gold and len(order) == len(set(order)), 'INVALID_METRIC_POPULATION')
    flags = [int(x in gold) for x in order[:50]]
    out = {}
    for k in (5, 10):
        hits = sum(flags[:k])
        idcg = math.fsum(1 / math.log2(i + 2) for i in range(min(k, len(gold))))
        out.update({f'hit_at_{k}': float(hits > 0), f'precision_at_{k}': hits / k,
                    f'recall_at_{k}': hits / len(gold),
                    f'ndcg_at_{k}': math.fsum(v / math.log2(i + 2) for i, v in enumerate(flags[:k])) / idcg})
    first = next((i + 1 for i, v in enumerate(flags) if v), None)
    out['reciprocal_rank'] = 1 / first if first else 0.0
    return out

def mean_metric(values):
    rows = list(values)
    require(bool(rows), 'EMPTY_METRIC_COHORT')
    return {k: math.fsum(r[k] for r in rows) / len(rows) for k in sorted(rows[0])}

def rank_ids(row, count):
    c = row['candidates']
    require(len(c) == count and [r['rank'] for r in c] == list(range(1, count + 1)), 'RANK_SEQUENCE_OR_DEPTH_CONFLICT')
    ids = [r['chunk_id'] for r in c]
    require(len(ids) == len(set(ids)), 'DUPLICATE_RANKED_CHUNK')
    return ids

def run_action(number, fn, dependencies=()):
    blocked = [d for d in dependencies if STATE['actions'].get(d, {}).get('status') != 'COMPLETE_IN_SCOPE']
    if blocked:
        out = {'status': 'BLOCKED_BY_DEPENDENCY', 'dependencies': blocked}
    else:
        try:
            out = fn()
            out['status'] = 'COMPLETE_IN_SCOPE'
        except InputGap as error:
            out = {'status': 'BLOCKED_INPUT_OR_CONTRACT', 'reason': str(error)}
        except Exception as error:
            # Real implementation errors remain distinguishable from missing input.
            out = {'status': 'FAILED_IMPLEMENTATION', 'error_type': type(error).__name__}
            STATE['errors'].append({'action': number, 'error_type': type(error).__name__})
    STATE['actions'][number] = out
    print(json.dumps({'action': number, **out, 'new_api_calls': 0}, ensure_ascii=False, indent=2))
    return out

def components(dev):
    parent = {q: q for q in dev['ids']}
    def find(x):
        while parent[x] != x:
            parent[x] = parent[parent[x]]; x = parent[x]
        return x
    seen = {}
    for q in dev['ids']:
        for cid in dev['gold'][q]:
            sid = dev['chunks'][cid].get('source_id')
            require(isinstance(sid, str) and sid.strip(), 'GOLD_SOURCE_PROVENANCE_REQUIRED')
            if sid in seen:
                a, b = find(q), find(seen[sid])
                parent[max(a, b)] = min(a, b)
            else:
                seen[sid] = q
    return {q: find(q) for q in dev['ids']}

def partition(component_map):
    labels = {c: ('validation' if int(digest((CONFIG['seed'] + c).encode())[:16], 16) % 100
                    < CONFIG['validation_percent'] else 'tuning') for c in set(component_map.values())}
    out = {name: sorted(q for q, c in component_map.items() if labels[c] == name)
           for name in ('tuning', 'validation')}
    for name, ids in out.items():
        require(len({component_map[q] for q in ids}) >= CONFIG['minimum_components_per_partition'],
                'INSUFFICIENT_SOURCE_COMPONENTS:' + name)
    return out

def eligible(candidate, reference):
    return all(candidate[k] >= reference[k] - 1e-12 for k in ('recall_at_10', 'hit_at_10')) and candidate['ndcg_at_10'] > reference['ndcg_at_10'] + 1e-12


def load_local_predictor():
 if LAB['predict'] is not None:return LAB['predict']
 print(json.dumps({'stage':'INITIALIZE_PINNED_LOCAL_ONNX','downloads':False}),flush=True)
 try:
  import numpy as np
  import onnxruntime as ort
  from transformers import AutoTokenizer
 except ImportError as e:raise InputGap('LOCAL_RUNTIME_PACKAGE_UNAVAILABLE:'+str(e.name)) from e
 require(ort.__version__==MODEL['onnxruntime_version'],'PINNED_ONNXRUNTIME_VERSION_REQUIRED')
 cache_root=Path(os.environ.get('HF_HUB_CACHE') or str(Path(os.environ.get('HF_HOME',str(Path.home()/'.cache/huggingface')))/'hub'))
 snapshot=cache_root/('models--'+MODEL['model_id'].replace('/','--'))/'snapshots'/MODEL['revision']
 model=snapshot/'onnx/model_quint8_avx2.onnx'
 require(model.is_file(),'PINNED_LOCAL_ONNX_NOT_CACHED_NO_DOWNLOAD')
 require(digest(model.read_bytes())==MODEL['model_sha256'],'MODEL_BYTES_IDENTITY_CONFLICT')
 tokenizer=AutoTokenizer.from_pretrained(str(snapshot),local_files_only=True,trust_remote_code=False)
 opts=ort.SessionOptions();opts.intra_op_num_threads=4;opts.inter_op_num_threads=1
 opts.execution_mode=ort.ExecutionMode.ORT_SEQUENTIAL;opts.graph_optimization_level=ort.GraphOptimizationLevel.ORT_ENABLE_ALL
 session=ort.InferenceSession(str(model),sess_options=opts,providers=['CPUExecutionProvider'])
 names={x.name for x in session.get_inputs()}
 require({'input_ids','attention_mask'}.issubset(names),'ONNX_INPUT_CONTRACT_CONFLICT')
 def predict(q,ids):
  require(0<len(ids)<=4,'BATCH_SIZE_CONFLICT')
  encoded=tokenizer([LAB['questions'][q]]*len(ids),[LAB['dev']['chunks'][c]['content'] for c in ids],
    padding=True,truncation=True,max_length=512,return_tensors='np')
  feed={name:np.asarray(encoded[name],dtype=np.int64) for name in names if name in encoded}
  require(set(feed)==names,'TOKENIZER_MODEL_INPUT_CONFLICT')
  values=np.asarray(session.run(None,feed)[0],dtype=np.float64).reshape(-1)
  require(values.shape==(len(ids),) and bool(np.isfinite(values).all()),'INVALID_MODEL_SCORES')
  return [float(v) for v in values]
 # Recheck current CPU runtime parity before mixing cached and new model scores.
 q=LAB['pilot'][0];sample=list(LAB['legacy'][q])[:4]
 observed=predict(q,sample);LAB['parity_pairs']+=len(sample);delta=max(abs(v-LAB['legacy'][q][cid]) for v,cid in zip(observed,sample))
 require(delta<=1e-4,'LOCAL_SCORE_PARITY_FAILURE_NO_NEW_SCORE_MIXING')
 LAB['guard']=save('runtime_parity',{'model_contract_sha256':LAB['model_contract']['sha256'],
  'model_sha256':MODEL['model_sha256'],'onnxruntime_version':ort.__version__,'sample_max_delta':delta,
  'sample_query':q,'sample_chunks':sample,'scope':'FIRST4_SAMPLE_ONLY_NOT_FULL_NUMERICAL_EQUIVALENCE'})
 LAB['predict']=predict
 return predict

def score_cache(q):
 identity={'schema_version':'NB22E_PAIR_CACHE_V1','model_contract_sha256':LAB['model_contract']['sha256'],
  'candidate_id':q,'question_sha256':digest(LAB['questions'][q].encode())}
 folder=path_for('artifacts/phase2/quality_improvement/NB22E.pair_cache_v1')
 folder.mkdir(parents=True,exist_ok=True);p=folder/(digest(canonical(identity))+'.json')
 if p.is_file():
  before=p.stat();raw=p.read_bytes();after=p.stat();require((before.st_size,before.st_mtime_ns)==(after.st_size,after.st_mtime_ns),'PAIR_CACHE_CHANGED_DURING_READ')
  record=decode(raw);require(record['identity']==identity and record['pairs_sha256']==digest(canonical(record['pairs'])),'PAIR_CACHE_IDENTITY_OR_CHECKSUM_CONFLICT')
  guard=read_json(record['runtime_guard'])
  require(guard['model_contract_sha256']==identity['model_contract_sha256'] and guard['model_sha256']==MODEL['model_sha256'] and 0<=guard['sample_max_delta']<=1e-4,'PAIR_CACHE_RUNTIME_GUARD_CONFLICT')
  for cid,item in record['pairs'].items():
   require(cid in LAB['dev']['chunks'] and item['content_sha256']==digest(LAB['dev']['chunks'][cid]['content'].encode()) and type(item['score']) in (int,float) and math.isfinite(item['score']),'PAIR_CACHE_TEXT_OR_VALUE_CONFLICT')
 else:record={'identity':identity,'pairs':{}}
 return p,record

def persist_pairs(path,record):
 record['pairs_sha256']=digest(canonical(record['pairs']));raw=canonical(record)
 fd,name=tempfile.mkstemp(dir=path.parent,prefix='.nb22e-score-')
 try:
  with os.fdopen(fd,'wb') as f:f.write(raw);f.flush();os.fsync(f.fileno())
  os.replace(name,path)
 finally:Path(name).unlink(missing_ok=True)
 require(path.read_bytes()==raw and decode(path.read_bytes())==record,'PAIR_SCORE_CHECKPOINT_RELOAD_FAILURE')

def score_order(q,order):
 path,record=score_cache(q)
 scores=dict(LAB['legacy'][q]);scores.update({c:x['score'] for c,x in record['pairs'].items()})
 missing=[c for c in order if c not in scores]
 for start in range(0,len(missing),4):
  batch=missing[start:start+4]
  require(LAB['new_pairs']+len(batch)<=CONFIG['maximum_new_pairs_per_run'],'NEW_PAIR_BUDGET_REACHED_RESUME_CACHE_SAFE')
  require(time.monotonic()-LAB['started']<CONFIG['maximum_local_seconds_per_run'],'LOCAL_TIME_CAP_REACHED_RESUME_CACHE_SAFE')
  predict=load_local_predictor();t=time.monotonic();values=predict(q,batch)
  LAB['new_inference_seconds']+=time.monotonic()-t
  require(len(values)==len(batch) and all(type(x) in (int,float) and math.isfinite(x) for x in values),'INVALID_NEW_PAIR_VALUES')
  for c,v in zip(batch,values):
   record['pairs'][c]={'score':v,'content_sha256':digest(LAB['dev']['chunks'][c]['content'].encode())};scores[c]=v
  record['runtime_guard']=LAB['guard'];persist_pairs(path,record)
  LAB['new_pairs']+=len(batch);STATE['model_pairs']+=len(batch)
  if time.monotonic()-LAB['last_progress']>=20:
   print(json.dumps({'stage':'LOCAL_PAIR_SCORING','candidate_id':q,'cached_or_legacy_pairs_for_query':len(scores),
    'required_for_route':len(order),'new_pairs_this_run':LAB['new_pairs'],'parity_check_pairs':LAB['parity_pairs'],
    'elapsed_seconds':round(time.monotonic()-LAB['started'],1)}),flush=True);LAB['last_progress']=time.monotonic()
 return scores

def run_route(spec):
 name=spec['name'];started=time.monotonic();before_pairs=LAB['new_pairs'];before_inference=LAB['new_inference_seconds']
 if not LAB['ready']:
  result={'status':'BLOCKED_SETUP','issues':LAB['issues'],'specification':spec}
 elif spec['mode']!='baseline' and not LAB['model_ready']:
  result={'status':'BLOCKED_MODEL_PREREQUISITE','issues':LAB['model_issues'],'specification':spec}
 else:
  if LAB['started'] is None:LAB['started']=time.monotonic()
  values={};orders={};pool_ceiling={};held=[]
  for q in LAB['pilot']:
   try:
    if spec['mode']=='baseline':
     order=LAB['incumbent'][q];pool=order
    else:
     pool=LAB['pools'][q][:spec['depth']];scores=score_order(q,pool)
     order=sorted(pool,key=lambda c:(-scores[c],c))
     if spec['mode']=='fusion':
      rr={c:i+1 for i,c in enumerate(order)};pr={c:i+1 for i,c in enumerate(pool)};w=CONFIG['reranker_weight'];k=CONFIG['rrf_k']
      order=sorted(pool,key=lambda c:(-(w/(k+rr[c])+(1-w)/(k+pr[c])),c))
    values[q]=metrics(order,LAB['dev']['gold'][q]);orders[q]=order
    pool_ceiling[q]=min(10,len(set(pool)&LAB['dev']['gold'][q]))/len(LAB['dev']['gold'][q])
   except InputGap as e:held.append({'candidate_id':q,'reason':str(e)})
   except Exception as e:held.append({'candidate_id':q,'reason':'RUNTIME_FAILURE:'+type(e).__name__})
   print(json.dumps({'route':name,'completed':len(values),'held':len(held),'total':len(LAB['pilot']),
    'new_pairs_this_route':LAB['new_pairs']-before_pairs,'elapsed_seconds':round(time.monotonic()-started,1)},ensure_ascii=False),flush=True)
   if held:break  # shared runtime/limit error; preserve partial route and continue independent route cells.
  complete=set(values)==set(LAB['pilot'])
  artifact=save('route_'+name,{'specification':spec,'plan':LAB['plan'],'metrics_by_query':values,'orders':orders,
   'pool_ceiling_by_query':pool_ceiling,'held':held,'complete':complete})
  result={'status':'COMPLETE' if complete else 'PARTIAL_HELD_NOT_ELIGIBLE','specification':spec,
   'query_count':len(values),'expected_queries':len(LAB['pilot']),'metrics':mean_metric(values.values()) if complete else None,
   'pool_ceiling':statistics.fmean(pool_ceiling.values()) if complete else None,
   'new_pairs':LAB['new_pairs']-before_pairs,'route_wall_seconds':time.monotonic()-started,
   'inference_seconds_new_pairs':LAB['new_inference_seconds']-before_inference,
   'timing_scope':'INCREMENTAL_SHARED_CACHE_NOT_FAIR_STANDALONE_MODEL_LATENCY',
   'held':held,'artifact':artifact,'values':values}
 LAB['results'][name]=result
 print(json.dumps({'route':name,**{k:v for k,v in result.items() if k!='values'}},ensure_ascii=False,indent=2))
 return result

def select_pilot(ids,golden,cmap):
 strata={source:[q for q in ids if set(golden[q]['source_types'])=={source}]
         for source in ('book','transcript','website')}
 require(all(strata.values()),'SOURCE_STRATUM_WITHOUT_ANY_DEVELOPMENT_QUERY')
 target=min(CONFIG['pilot_per_source_type'],min(map(len,strata.values())))
 pilot=[];audit={};used_components=set()
 for source,qs in strata.items():
  groups=collections.defaultdict(list)
  for q in qs:groups[cmap[q]].append(q)
  for group in groups.values():group.sort(key=lambda q:(digest((CONFIG['seed']+q).encode()),q))
  order=sorted(groups,key=lambda c:(c in used_components,digest((CONFIG['seed']+source+c).encode()),c))
  chosen=[];round_index=0
  while len(chosen)<target:
   advanced=False
   for component in order:
    if round_index<len(groups[component]):
     chosen.append(groups[component][round_index]);advanced=True
     if len(chosen)==target:break
   require(advanced,'INTERNAL_SAMPLER_EXHAUSTED')
   round_index+=1
  pilot.extend(chosen);used_components.update(cmap[q] for q in chosen)
  audit[source]={'available_queries':len(qs),'available_components':len(groups),'selected_queries':len(chosen),
   'selected_components':len({cmap[q] for q in chosen}),'repeated_component_queries':len(chosen)-len({cmap[q] for q in chosen})}
 require(len(pilot)==len(set(pilot))==3*target,'PILOT_SAMPLING_COUNT_CONFLICT')
 return pilot,{'strata':audit,'target_per_source':target,'total_queries':len(pilot),
  'distinct_components_total':len({cmap[q] for q in pilot}),
  'all_queries_source_component_independent':len({cmap[q] for q in pilot})==len(pilot),
  'scope':'DESCRIPTIVE_DEVELOPMENT_PILOT_NOT_INDEPENDENT_STATISTICAL_ACCEPTANCE',
  'reduced_for_real_query_availability':target<CONFIG['pilot_per_source_type']}

def prepare_lab():
 require(SAVED_CODE_VALID,'SAVED_NOTEBOOK_CODE_IDENTITY_REQUIRED')
 verified=read_json(PINS['C056']);previous=read_json(PINS['D_final'])
 require(verified['independent_recomputation_complete'] and previous['pool_ready_for_bounded_reranker_pilot'],'VERIFIED_EXPANSION_REQUIRED')
 bindings=verified['input_file_receipts'];chunks=keyed(read_rows(bindings['chunks']),'chunk_id')
 golden=keyed(read_rows(bindings['golden']),'candidate_id');hybrid=keyed(read_rows(bindings['hybrid']),'candidate_id')
 require(len(chunks)==3488 and len(golden)==3813 and set(golden)==set(hybrid),'POPULATION_CONFLICT')
 loc=collections.defaultdict(set);types=collections.defaultdict(set)
 for cid,c in chunks.items():
  for m in c['member_locators']:
   k=canonical(m['locator']);loc[k].add(cid);types[k].add(c['source_type'])
 gold={}
 for q,row in golden.items():
  g=set()
  for e in row['evidence_locators']:
   k=canonical(e['source_locator']);require(bool(loc[k]) and bool(types[k]&set(row['source_types'])),'GOLD_MAPPING_CONFLICT');g.update(loc[k])
  require(bool(g),'EMPTY_GOLD');gold[q]=g
 md=hashlib.sha256()
 for q in sorted(gold):
  raw=canonical({'candidate_id':q,'relevant_chunk_ids':sorted(gold[q])});md.update(len(raw).to_bytes(8,'big'));md.update(raw)
 require(md.hexdigest()==verified['gold_mapping']['recomputed_sha256'],'LOGICAL_GOLD_CONFLICT')
 ids=sorted(q for q in hybrid if hybrid[q]['split']=='development');require(len(ids)==2664,'DEVELOPMENT_COUNT_CONFLICT')
 dev={'ids':ids,'chunks':chunks,'gold':{q:gold[q] for q in ids}}
 cmap=components(dev)
 pilot,sampling=select_pilot(ids,golden,cmap)
 pool_binding=previous['pool_artifact'];ranking_binding=previous['retained_rankings']
 # Stream large artifacts, validate full hashes, retain only pilot orders.
 def pilot_rows(binding):
  p=path_for(binding['path']);require(p.is_file(),'REQUIRED_FILE_NOT_AVAILABLE:'+binding['path']);before=p.stat();h=hashlib.sha256();out={}
  with p.open('rb') as f:
   for raw in f:
    h.update(raw);row=decode(raw)
    if row.get('candidate_id') in pilot:
     require(row['candidate_id'] not in out,'DUPLICATE_PILOT_ROW');out[row['candidate_id']]=row
  after=p.stat();require((before.st_size,before.st_mtime_ns)==(after.st_size,after.st_mtime_ns) and h.hexdigest()==binding_hash(binding),'PILOT_ARTIFACT_HASH_OR_MUTATION_CONFLICT')
  require(set(out)==set(pilot),'PILOT_COHORT_CONFLICT');STATE['inputs'][binding['path']]=binding;return out
 pool_rows=pilot_rows(pool_binding);rank_rows=pilot_rows(ranking_binding);reranker=pilot_rows(bindings['reranker'])
 pools={};incumbent={};legacy={}
 for q in pilot:
  pool=pool_rows[q]['candidate_chunk_ids'];order=rank_rows[q]['ranked_chunk_ids']
  require(pool_rows[q]['split']==rank_rows[q]['split']==reranker[q]['split']=='development','PILOT_SPLIT_CONFLICT')
  require(len(pool)<=800 and len(pool)==len(set(pool)) and set(rank_ids(hybrid[q],100)).issubset(pool) and set(pool).issubset(chunks),'EXPANDED_POOL_CONFLICT')
  require(set(order).issubset(chunks) and all(abs(v-rank_rows[q]['metrics'][k])<=1e-12 for k,v in metrics(order,gold[q]).items()),'BASELINE_REPLAY_CONFLICT')
  pools[q]=pool;incumbent[q]=order;legacy[q]={c['chunk_id']:float(c['reranker_score']) for c in reranker[q]['candidates']}
 model_error=None;found=set();contract=None
 try:
  inventory=read_json(PINS['B_cache_inventory']);contract=inventory['inference_contract'];spec=contract['specification']
  require(digest(canonical(spec))==contract['sha256'],'LEGACY_CONTRACT_CHECKSUM_CONFLICT')
  for key in MODEL:
   require(spec.get(key)==MODEL[key],'LEGACY_MODEL_CONTRACT_CONFLICT:'+key)
  require(spec['baseline_reranker_file_bytes_sha256']==binding_hash(bindings['reranker']),'LEGACY_BASELINE_SCORE_BINDING_CONFLICT')
  require(spec['execution_provider']=='CPUExecutionProvider' and spec['threads']=={'intra_op':4,'inter_op':1},'LEGACY_EXECUTION_PROFILE_CONFLICT')
  found=set()
  for binding in inventory['files']:
   if not binding['path'].startswith('artifacts/phase2/quality_improvement/NB22B.depth100/'):continue
   # Guards and records share a directory; records have the per-query identity.
   record=read_json(binding);identity=record.get('identity',{});q=identity.get('candidate_id')
   if q not in pilot:continue
   require(q not in found and identity['inference_contract_sha256']==contract['sha256'],'LEGACY_QUERY_CONTRACT_CONFLICT')
   question=golden[q]['evaluation_content']['question'].strip()
   require(identity['question_sha256']==digest(question.encode()),'LEGACY_QUESTION_BINDING_CONFLICT')
   require(digest(canonical(record['scores']))==record['score_values_sha256'],'LEGACY_SCORE_CHECKSUM_CONFLICT')
   missing=identity['ordered_missing_chunks'];require(len(missing)==len(record['scores']),'LEGACY_SCORE_COUNT_CONFLICT')
   guard=read_json(record['runtime_validation_receipt'])
   require(guard['inference_contract_sha256']==contract['sha256'] and guard['model_sha256']==MODEL['model_sha256'] and 0<=guard['sample_score_max_delta']<=1e-4,'LEGACY_RUNTIME_GUARD_CONFLICT')
   for item,value in zip(missing,record['scores']):
    cid=item['chunk_id'];require(cid in chunks and item['content_sha256']==digest(chunks[cid]['content'].encode()) and type(value) in (int,float) and math.isfinite(value),'LEGACY_PAIR_BINDING_CONFLICT');legacy[q][cid]=value
   require(set(legacy[q])==set(rank_ids(hybrid[q],100)),'LEGACY100_POOL_CONFLICT');found.add(q)
  require(found==set(pilot),'COMPLETE_PRIOR100_SCORE_CACHE_REQUIRED_NO_RECOMPUTE_FALLBACK')
 except InputGap as e:
  model_error=str(e)
 except (KeyError,TypeError,ValueError) as e:
  model_error='LEGACY_SCORE_SCHEMA_CONFLICT:'+type(e).__name__
 LAB['model_ready']=model_error is None
 if model_error:LAB['model_issues'].append(model_error)
 LAB.update(ready=True,dev=dev,pilot=pilot,questions={q:golden[q]['evaluation_content']['question'].strip() for q in pilot},
  pools=pools,incumbent=incumbent,legacy=legacy,model_contract=contract,source_types={q:golden[q]['source_types'] for q in pilot},sampling=sampling)
 plan={'schema_version':CONFIG['version'],'producer_code_sha256':NOTEBOOK_CODE_HASH,'configuration':CONFIG,'pilot':pilot,
  'pilot_selection':'balanced source strata, component round-robin, fill within existing components when scarce; no outcome-based sampling',
  'sampling':sampling,'model_routes_ready':LAB['model_ready'],'model_prerequisite_issues':LAB['model_issues'],
  'source_inputs':STATE['inputs'],'model_contract':contract,
  'routes':ROUTES,'selection':'Exploratory pilot: highest nDCG@10 among full identical-cohort routes with Recall@10 and Hit@10 nondecreasing vs NB22D, strict nDCG gain. Tie route name. No automatic full execution.',
  'scope':'SMALL_DEVELOPMENT_PILOT_ALREADY_EXPOSED_NOT_FINAL_HOLDOUT','test_evaluation':False,'release_acceptance_granted':False,'api_calls':0}
 LAB['plan']=save('A01.frozen_pilot_plan',plan)
 return {'pilot_queries':len(pilot),'source_components':len({cmap[q] for q in pilot}),'model_downloads':False,'legacy_queries_reused':len(found),'model_routes_ready':LAB['model_ready'],'model_issues':LAB['model_issues'],'sampling':sampling,'plan':LAB['plan']}
try:
 print(json.dumps({'setup':'COMPLETE',**prepare_lab()},ensure_ascii=False,indent=2))
except InputGap as e:
 LAB['ready']=False;LAB['issues'].append(str(e));print(json.dumps({'setup':'BLOCKED','reason':str(e)},ensure_ascii=False,indent=2))

except Exception as e:
 LAB['ready']=False;LAB['issues'].append('SETUP_IMPLEMENTATION_OR_SCHEMA_FAILURE:'+type(e).__name__)
 print(json.dumps({'setup':'FAILED_SETUP','error_type':type(e).__name__,'no_inference_started':True},ensure_ascii=False,indent=2))
```

```python
# NB22E-C008 — Compare identical complete pilot cohorts; no full-run dispatch.
rows=[];baseline=LAB['results'].get('baseline_D',{});reference=baseline.get('metrics');eligible_rows=[]
for spec in ROUTES:
 name=spec['name'];r=LAB['results'].get(name,{'status':'NOT_RUN'});m=r.get('metrics')
 row={'route':name,'status':r['status'],'query_count':r.get('query_count',0),'metrics':m,
  'new_pairs':r.get('new_pairs',0),'seconds_incremental':r.get('route_wall_seconds'),
  'pool_ceiling_not_actual_top10':r.get('pool_ceiling')}
 if m is not None:
  row['acceptance']={k:{'threshold':v,'status':'PASS' if m[k]>=v else 'FAIL','mark':'✅' if m[k]>=v else ''} for k,v in CONFIG['thresholds'].items()}
  row['hit_absolute_threshold']=None
  row['eligible_for_larger_development_pilot']=name!='baseline_D' and reference is not None and eligible(m,reference)
  if row['eligible_for_larger_development_pilot']:eligible_rows.append(row)
 rows.append(row)
winner=sorted(eligible_rows,key=lambda r:(-r['metrics']['ndcg_at_10'],r['route']))[0]['route'] if eligible_rows else None
print('\n| مسیر | وضعیت | سؤال | Recall@10؛ حد 85% | قبول | nDCG@10؛ حد 75% | قبول | Hit@10؛ حد تعیین نشده | جفت تازه | ثانیهٔ افزایشی |')
print('|---|---|---:|---:|:---:|---:|:---:|---:|---:|---:|')
for row in rows:
 m=row['metrics'];a=row.get('acceptance',{})
 print('| '+row['route']+' | '+row['status']+' | '+str(row['query_count'])+' | '+
  (' | '.join([f"{m['recall_at_10']*100:.2f}%",a['recall_at_10']['mark'],f"{m['ndcg_at_10']*100:.2f}%",a['ndcg_at_10']['mark'],f"{m['hit_at_10']*100:.2f}%"]) if m else '— | | — | | —')+
  ' | '+str(row['new_pairs'])+' | '+(f"{row['seconds_incremental']:.1f}" if row['seconds_incremental'] is not None else '—')+' |')
summary={'schema_version':'NB22E_LAB_V2_FINAL','pilot_complete_for_all_routes':all(r['status']=='COMPLETE' for r in rows),
 'setup_ready':LAB['ready'],'issues':LAB['issues'],'model_issues':LAB['model_issues'],'sampling':LAB.get('sampling'),'routes':rows,'exploratory_winner':winner,
 'next_step':'REVIEW_PILOT_QUALITY_AND_TOTAL_COMPUTE_BEFORE_LARGER_DEVELOPMENT_RUN' if winner else 'KEEP_NB22D_AND_REVIEW_NEGATIVE_OR_HELD_ROUTES',
 'new_cache_miss_pairs':LAB['new_pairs'],'runtime_parity_check_pairs':LAB['parity_pairs'],
 'total_new_model_pair_evaluations':LAB['new_pairs']+LAB['parity_pairs'],'new_api_calls':0,'downloads':False,'test_evaluation':False,
 'release_acceptance_granted':False,'pilot_is_not_independent_acceptance_or_statistical_proof':True,
 'runtime_note':'Shared caches make later routes cheaper. Do not compare incremental route time as standalone speed. No full-cohort duration extrapolation is claimed from cache-hit timings.',
 'source_inputs':STATE['inputs'],'producer_code_sha256':NOTEBOOK_CODE_HASH}
if LAB['ready']:
 summary['plan']=LAB['plan']
 details={};component_map=components(LAB['dev'])
 baseline_values=baseline.get('values',{})
 for name,r in LAB['results'].items():
  if r['status']!='COMPLETE':continue
  by_source={}
  for source in ('book','transcript','website'):
   qs=[q for q in LAB['pilot'] if set(LAB['source_types'][q])=={source}]
   by_source[source]={'queries':len(qs),'source_components':len({component_map[q] for q in qs}),
    'metrics':mean_metric(r['values'][q] for q in qs)}
  cases=[]
  for q in LAB['pilot']:
   if q not in baseline_values:continue
   delta={k:r['values'][q][k]-baseline_values[q][k] for k in ('recall_at_10','ndcg_at_10','hit_at_10')}
   cases.append({'candidate_id':q,'source_types':LAB['source_types'][q],'delta':delta})
  details[name]={'source_slices':by_source,'paired_query_changes':cases}
 summary['paired_details']=details
 summary['report']=save('A02.lab_comparison',dict(summary))
print(json.dumps(summary,ensure_ascii=False,indent=2))
```

## اعتبارسنجی و تغییرات v2

خطای مشاهده‌شدهٔ v1، الزام چهار مؤلفهٔ مستقل کتاب، با نمونه‌گیری متوازن و گردش بین مؤلفه‌های موجود اصلاح شد. اگر تعداد مؤلفه‌ها کم باشد، سؤال‌های متفاوت از همان مؤلفه انتخاب و وابستگی در گزارش ثبت می‌شود؛ حد پذیرش و Gold تغییر نمی‌کند.

۳۳ کنترل نرم‌افزاری پاس شد؛ از جمله اجرای کامل با یک مؤلفهٔ کتاب، قالب واقعی source_types گلدن و شناسه‌های Development، کاهش متوازن تعداد نمونه، نبود واقعی یک نوع منبع، استقلال گزارش مرجع از پیش‌نیاز مدل، پنج مسیر، کش و Resume، حد محاسبات، کش خراب، تغییر متن، مقایسهٔ جمعیت یکسان و گزارش تفکیکی. آزمون inference از پیش‌بینی‌گر مصنوعی استفاده کرده است؛ مدل واقعی ONNX و کیفیت پروژه در این محیط اجرا/اندازه‌گیری نشده‌اند.

نسخهٔ v1 و شواهد گذشته حفظ شده‌اند. خروجی جدید در پوشهٔ v2 نوشته می‌شود؛ کش جفت‌های معتبر v1 قابل مصرف است. اجرای بدون مانع روی هر محیطی تضمین نیست؛ مانع باید با دلیل روشن گزارش شود و هیچ مسیر ناقص PASS نام نمی‌گیرد.

```python
# NB22E-C009 — آزمایش یکپارچه با نسخهٔ مستقل امتیازهای محلی
require(LAB.get("ready"), "Bootstrap نسخه v2 باید COMPLETE باشد.")
require(LAB.get("model_ready"), "پیش‌نیاز مدل در Bootstrap آماده نیست.")

# نسخهٔ امتیاز جدید؛ هیچ امتیاز قدیمی وارد مسیرهای مدل نمی‌شود.
origin = LAB.setdefault("coherent_origin_contract", LAB["model_contract"])
fresh_spec = {
    "schema_version": "NB22E_COHERENT_LOCAL_SCORE_V1",
    "parent_model_contract_sha256": origin["sha256"],
    "model": dict(MODEL),
    "score_policy": "ALL_POOL_PAIRS_SCORED_LOCALLY_NO_LEGACY_SCORE_MIXING",
    "preprocessing": "paired_tokenizer_dynamic_padding_truncation512_batch4",
    "runtime": {
        "python": platform.python_version(),
        "system": platform.system(),
        "machine": platform.machine(),
        "provider": "CPUExecutionProvider",
        "intra_op_threads": 4,
        "inter_op_threads": 1,
    },
}
LAB["model_contract"] = {
    "specification": fresh_spec,
    "sha256": digest(canonical(fresh_spec)),
}
LAB.update(
    legacy={q: {} for q in LAB["pilot"]},
    predict=None,
    results={},
    started=None,
    new_pairs=0,
    parity_pairs=0,
    new_inference_seconds=0.0,
    last_progress=0.0,
    fresh_runtime_failure=None,
)
LAB.setdefault("prior_plan_before_coherent", LAB["plan"])
LAB["plan"] = save("C009.coherent_score_plan", {
    "version": "NB22E_C009_COHERENT_LOCAL_SCORE_V1",
    "model_contract": LAB["model_contract"],
    "pilot": LAB["pilot"],
    "sampling": LAB["sampling"],
    "routes": ROUTES,
    "configuration": CONFIG,
    "prior_plan": LAB["prior_plan_before_coherent"],
    "reason": (
        "Prior/new numeric equivalence was not established; "
        "use a separate full-pool score version."
    ),
    "maximum_new_scoring_pairs": sum(
        len(LAB["pools"][q][:800]) for q in LAB["pilot"]
    ),
    "runtime_validation": (
        "Repeatability of identical new batches; "
        "not equivalence to old scores."
    ),
    "api_calls": 0,
    "downloads": False,
    "test_evaluation": False,
    "release_acceptance_granted": False,
})


def build_coherent_predictor():
    import numpy as np
    import onnxruntime as ort
    from transformers import AutoTokenizer

    require(
        ort.__version__ == MODEL["onnxruntime_version"],
        "PINNED_ONNX_VERSION_REQUIRED",
    )
    cache_root = Path(
        os.environ.get("HF_HUB_CACHE")
        or str(
            Path(os.environ.get(
                "HF_HOME", str(Path.home() / ".cache/huggingface")
            )) / "hub"
        )
    )
    snapshot = (
        cache_root
        / ("models--" + MODEL["model_id"].replace("/", "--"))
        / "snapshots"
        / MODEL["revision"]
    )
    model = snapshot / "onnx/model_quint8_avx2.onnx"
    require(model.is_file(), "PINNED_LOCAL_MODEL_NOT_CACHED_NO_DOWNLOAD")
    require(
        digest(model.read_bytes()) == MODEL["model_sha256"],
        "MODEL_BYTE_HASH_CONFLICT",
    )
    tokenizer = AutoTokenizer.from_pretrained(
        str(snapshot),
        local_files_only=True,
        trust_remote_code=False,
    )
    options = ort.SessionOptions()
    options.intra_op_num_threads = 4
    options.inter_op_num_threads = 1
    options.execution_mode = ort.ExecutionMode.ORT_SEQUENTIAL
    options.graph_optimization_level = ort.GraphOptimizationLevel.ORT_ENABLE_ALL

    session = ort.InferenceSession(
        str(model),
        sess_options=options,
        providers=["CPUExecutionProvider"],
    )
    names = {x.name for x in session.get_inputs()}
    require(
        {"input_ids", "attention_mask"}.issubset(names),
        "MODEL_INPUT_CONTRACT_CONFLICT",
    )

    def predict(q, ids):
        require(0 < len(ids) <= 4, "BATCH_SIZE_CONFLICT")
        encoded = tokenizer(
            [LAB["questions"][q]] * len(ids),
            [LAB["dev"]["chunks"][c]["content"] for c in ids],
            padding=True,
            truncation=True,
            max_length=512,
            return_tensors="np",
        )
        feed = {
            name: np.asarray(encoded[name], dtype=np.int64)
            for name in names if name in encoded
        }
        require(set(feed) == names, "TOKENIZER_MODEL_INPUT_CONFLICT")
        values = np.asarray(
            session.run(None, feed)[0], dtype=np.float64
        ).reshape(-1)
        require(
            values.shape == (len(ids),) and bool(np.isfinite(values).all()),
            "INVALID_NEW_SCORES",
        )
        return [float(v) for v in values]

    # کنترل تکرارپذیری اجرای جدید؛ ادعای تطابق با امتیاز قدیمی نمی‌کند.
    q = LAB["pilot"][0]
    sample = LAB["pools"][q][:4]
    a, b = predict(q, sample), predict(q, sample)
    LAB["parity_pairs"] += 2 * len(sample)
    delta = max(abs(x - y) for x, y in zip(a, b))
    require(delta <= 1e-6, "NEW_RUNTIME_REPEATABILITY_FAILURE")

    LAB["guard"] = save("C009.new_runtime_repeatability", {
        "model_contract_sha256": LAB["model_contract"]["sha256"],
        "model_sha256": MODEL["model_sha256"],
        "sample_max_delta": delta,
        "validation_scope": (
            "IDENTICAL_NEW_BATCH_REPEATABILITY_ONLY_"
            "NOT_OLD_SCORE_EQUIVALENCE"
        ),
        "sample_query": q,
        "sample_chunks": sample,
    })
    LAB["predict"] = predict
    return predict


def load_local_predictor():
    if LAB["predict"] is not None:
        return LAB["predict"]
    require(
        LAB["fresh_runtime_failure"] is None,
        "LOCAL_RUNTIME_PREVIOUSLY_FAILED:"
        + str(LAB["fresh_runtime_failure"]),
    )
    try:
        return build_coherent_predictor()
    except Exception as error:
        reason = (
            str(error) if isinstance(error, InputGap)
            else type(error).__name__
        )
        LAB["fresh_runtime_failure"] = reason
        raise InputGap(
            "LOCAL_RUNTIME_INITIALIZATION_FAILED:" + reason
        ) from error


# تمام مسیرها، روی همان نمونه و با کش مشترکِ نسخهٔ جدید.
for route in ROUTES:
    run_route(route)

reference = LAB["results"]["baseline_D"].get("metrics")
rows, candidates = [], []

print("\nاین جدول فقط مربوط به پایلوت است؛ پذیرش کل محصول نیست.")
print(
    "| مسیر | وضعیت | Recall@10؛ هدف 85% | قبول | "
    "nDCG@10؛ هدف 75% | قبول | Hit@10؛ حد تعیین نشده |"
)
print("|---|---|---:|:---:|---:|:---:|---:|")

for route in ROUTES:
    name = route["name"]
    result = LAB["results"][name]
    m = result.get("metrics")
    row = {
        "route": name,
        **{k: v for k, v in result.items() if k != "values"},
    }
    if m is not None:
        recall_mark = "✅" if m["recall_at_10"] >= .85 else ""
        ndcg_mark = "✅" if m["ndcg_at_10"] >= .75 else ""
        print(
            f"| {name} | {result['status']} | "
            f"{100*m['recall_at_10']:.2f}% | {recall_mark} | "
            f"{100*m['ndcg_at_10']:.2f}% | {ndcg_mark} | "
            f"{100*m['hit_at_10']:.2f}% |"
        )
        if (
            name != "baseline_D"
            and reference is not None
            and eligible(m, reference)
        ):
            candidates.append(name)
    else:
        print(f"| {name} | {result['status']} | — | | — | | — |")
    rows.append(row)

winner = (
    sorted(
        candidates,
        key=lambda n: (
            -LAB["results"][n]["metrics"]["ndcg_at_10"], n
        ),
    )[0]
    if candidates else None
)
summary = {
    "version": "NB22E_C009_COHERENT_LOCAL_SCORE_V1",
    "all_routes_complete": all(r["status"] == "COMPLETE" for r in rows),
    "routes": rows,
    "exploratory_pilot_winner": winner,
    "new_cache_miss_pairs": LAB["new_pairs"],
    "repeatability_check_pairs": LAB["parity_pairs"],
    "plan": LAB["plan"],
    "model_contract": LAB["model_contract"],
    "scope": "SMALL_DEVELOPMENT_PILOT_NOT_RELEASE_ACCEPTANCE",
    "old_scores_mixed_into_new_model_routes": False,
    "api_calls": 0,
    "downloads": False,
    "test_evaluation": False,
    "release_acceptance_granted": False,
}
receipt = save("C009.coherent_comparison", summary)

print(json.dumps({
    "all_routes_complete": summary["all_routes_complete"],
    "exploratory_pilot_winner": winner,
    "new_pairs": LAB["new_pairs"],
    "repeatability_check_pairs": LAB["parity_pairs"],
    "report": receipt,
    "new_api_calls": 0,
}, ensure_ascii=False, indent=2))
```

### NB22E-C010 — اقدام ۳: ارزیابی گسترده‌تر روش منتخب

**مرز نتیجه:** این نمونه از Development قبلاً دیده‌شده است؛ وابستگی منابع گزارش می‌شود و آزمون مستقل نهایی یا پذیرش Release محسوب نمی‌شود.

**API و دانلود:** صفر. محاسبات مدل فقط روی دستگاه محلی انجام می‌شود.

**شرط پایان:** هر دو مسیر برای تمام ۶۰ سؤال کامل باشند و گزارش مقایسه ذخیره و بازخوانی شود. خروجی ناقص به کل نمونه تعمیم داده نمی‌شود.

```python
# NB22E-C010 — Action 3: frozen onnx400_fusion on 60 additional Development queries.
import unicodedata

require(LAB.get("ready") and LAB.get("model_ready"), "BOOTSTRAP_V2_REQUIRED")
require(
    LAB["model_contract"]["specification"].get("schema_version")
    == "NB22E_COHERENT_LOCAL_SCORE_V1",
    "RUN_C009_COHERENT_SETUP_FIRST",
)
require(
    CONFIG["rrf_k"] == 10 and CONFIG["reranker_weight"] == .75,
    "FROZEN_SELECTED_METHOD_PARAMETERS_CHANGED",
)

# Preserve the original pilot when this cell is rerun to resume work.
if "EXPAND60_ORIGIN" not in globals():
    require(len(LAB["pilot"]) == 12, "ORIGINAL_12_QUERY_PILOT_REQUIRED")
    EXPAND60_ORIGIN = {
        "pilot": list(LAB["pilot"]),
        "questions": dict(LAB["questions"]),
        "plan": LAB["plan"],
        "contract_sha256": LAB["model_contract"]["sha256"],
        "prior_new_pairs": LAB["new_pairs"],
    }

require(
    EXPAND60_ORIGIN["contract_sha256"] == LAB["model_contract"]["sha256"],
    "COHERENT_SCORE_VERSION_CHANGED",
)

# Keep already validated pair records in memory as well as on disk.
# Conflicting values or changed text remain errors.
if "_EXP60_STORAGE" not in globals():
    _EXP60_STORAGE = (score_cache, persist_pairs)
LAB.setdefault("verified_pair_memory", {})


def persist_pairs60(path, record):
    _EXP60_STORAGE[1](path, record)
    key = digest(canonical(record["identity"]))
    LAB["verified_pair_memory"][key] = decode(canonical(record))


def score_cache60(q):
    path, record = _EXP60_STORAGE[0](q)
    key = digest(canonical(record["identity"]))
    memo = LAB["verified_pair_memory"].get(key)
    if memo is not None:
        for cid, item in memo["pairs"].items():
            require(
                item["content_sha256"]
                == digest(LAB["dev"]["chunks"][cid]["content"].encode()),
                "MEMORY_PAIR_TEXT_CHANGED",
            )
            require(
                cid not in record["pairs"] or record["pairs"][cid] == item,
                "MEMORY_DISK_PAIR_SCORE_CONFLICT",
            )
        record["pairs"].update(memo["pairs"])
        record["runtime_guard"] = memo["runtime_guard"]
    return path, record


score_cache = score_cache60
persist_pairs = persist_pairs60

SEED60 = "NB22E_C010_EXPAND60_V1"
TARGET60 = 20

verified = read_json(PINS["C056"])
golden60 = keyed(
    read_rows(verified["input_file_receipts"]["golden"]), "candidate_id"
)
previous60 = read_json(PINS["D_final"])
component60 = components(LAB["dev"])


def question_key60(text):
    return " ".join(unicodedata.normalize("NFKC", text).split())


excluded60 = set(EXPAND60_ORIGIN["pilot"])
seen_text60 = {
    question_key60(q) for q in EXPAND60_ORIGIN["questions"].values()
}
selected60 = []
audit60 = {}
old_components60 = {component60[q] for q in excluded60}
used_components60 = set(old_components60)

# Sampling uses source/provenance, never model outcomes or evaluation metrics.
for source60 in ("book", "transcript", "website"):
    groups60 = collections.defaultdict(list)

    for q60 in LAB["dev"]["ids"]:
        row60 = golden60[q60]
        if q60 in excluded60 or set(row60["source_types"]) != {source60}:
            continue
        text60 = row60["evaluation_content"]["question"].strip()
        if question_key60(text60) not in seen_text60:
            groups60[component60[q60]].append(q60)

    for group60 in groups60.values():
        group60.sort(key=lambda q: (digest((SEED60 + q).encode()), q))

    order60 = sorted(
        groups60,
        key=lambda c: (
            c in used_components60,
            digest((SEED60 + source60 + c).encode()),
            c,
        ),
    )
    chosen60 = []
    level60 = 0

    while len(chosen60) < TARGET60:
        available60 = False
        for c60 in order60:
            if level60 >= len(groups60[c60]):
                continue
            available60 = True
            q60 = groups60[c60][level60]
            key60 = question_key60(
                golden60[q60]["evaluation_content"]["question"]
            )
            if key60 in seen_text60:
                continue
            chosen60.append(q60)
            seen_text60.add(key60)
            if len(chosen60) == TARGET60:
                break

        require(
            available60,
            "NOT_ENOUGH_DISTINCT_NEW_QUESTIONS:" + source60,
        )
        level60 += 1

    selected60.extend(chosen60)
    used_components60.update(component60[q] for q in chosen60)
    audit60[source60] = {
        "queries": len(chosen60),
        "components": len({component60[q] for q in chosen60}),
        "queries_sharing_components_with_pilot12": sum(
            component60[q] in old_components60 for q in chosen60
        ),
    }

require(
    len(selected60) == len(set(selected60)) == 60
    and not (set(selected60) & excluded60),
    "NEW60_COHORT_CONFLICT",
)


# Validate full source-artifact hashes; retain only the selected records.
def subset60(binding):
    p = path_for(binding["path"])
    require(
        p.is_file(),
        "REQUIRED_FILE_NOT_AVAILABLE:" + binding["path"],
    )
    before = p.stat()
    h = hashlib.sha256()
    out = {}

    with p.open("rb") as f:
        for raw in f:
            h.update(raw)
            row = decode(raw)
            q = row.get("candidate_id")
            if q in selected60:
                require(q not in out, "DUPLICATE_NEW60_QUERY")
                out[q] = row

    after = p.stat()
    require(
        (before.st_size, before.st_mtime_ns, before.st_ino)
        == (after.st_size, after.st_mtime_ns, after.st_ino)
        and h.hexdigest() == binding_hash(binding),
        "SOURCE_ARTIFACT_HASH_OR_MUTATION_CONFLICT",
    )
    require(
        set(out) == set(selected60),
        "SOURCE_ARTIFACT_NEW60_COVERAGE_CONFLICT",
    )
    return out


pools60 = subset60(previous60["pool_artifact"])
ranks60 = subset60(previous60["retained_rankings"])

for q60 in selected60:
    pool60 = pools60[q60]["candidate_chunk_ids"]
    rank60 = ranks60[q60]["ranked_chunk_ids"]
    require(
        pools60[q60]["split"] == ranks60[q60]["split"] == "development",
        "NEW60_SPLIT_CONFLICT",
    )
    require(
        0 < len(pool60) <= 800
        and len(pool60) == len(set(pool60))
        and set(pool60).issubset(LAB["dev"]["chunks"]),
        "NEW60_POOL_CONFLICT",
    )
    require(
        len(rank60) == len(set(rank60))
        and set(rank60).issubset(LAB["dev"]["chunks"]),
        "NEW60_BASELINE_RANK_CONFLICT",
    )
    require(
        all(
            abs(v - ranks60[q60]["metrics"][k]) <= 1e-12
            for k, v in metrics(rank60, LAB["dev"]["gold"][q60]).items()
        ),
        "NEW60_BASELINE_METRIC_REPLAY_CONFLICT",
    )

# Explicit local-compute budget for this larger pilot.
CONFIG["maximum_new_pairs_per_run"] = 24000
CONFIG["maximum_local_seconds_per_run"] = 7200

LAB.update(
    pilot=selected60,
    questions={
        q: golden60[q]["evaluation_content"]["question"].strip()
        for q in selected60
    },
    source_types={q: golden60[q]["source_types"] for q in selected60},
    pools={q: pools60[q]["candidate_chunk_ids"] for q in selected60},
    incumbent={q: ranks60[q]["ranked_chunk_ids"] for q in selected60},
    legacy={q: {} for q in selected60},
    results={},
    started=None,
    new_pairs=0,
    parity_pairs=0,
    new_inference_seconds=0.0,
    last_progress=0.0,
    fresh_runtime_failure=None,
)

LAB["plan"] = save("C010.expand60_frozen_plan", {
    "version": "NB22E_EXPAND60_V1",
    "producer_cell": "NB22E-C010",
    "seed": SEED60,
    "pilot_queries": selected60,
    "excluded_pilot12": sorted(excluded60),
    "sampling": audit60,
    "pilot12_plan": EXPAND60_ORIGIN["plan"],
    "model_contract": LAB["model_contract"],
    "selected_method": {
        "mode": "fusion", "depth": 400, "rrf_k": 10, "weight": .75
    },
    "selection_uses_outcomes": False,
    "source_artifacts": {
        "golden": verified["input_file_receipts"]["golden"],
        "pools": previous60["pool_artifact"],
        "baseline": previous60["retained_rankings"],
    },
    "maximum_new_pairs_this_run": 24000,
    "maximum_seconds_this_run": 7200,
    "new_pair_upper_bound": sum(
        min(400, len(LAB["pools"][q])) for q in selected60
    ),
    "scope": (
        "ADDITIONAL_PREVIOUSLY_EXPOSED_DEVELOPMENT_"
        "NOT_INDEPENDENT_FINAL_HOLDOUT"
    ),
    "api_calls": 0,
    "downloads": False,
    "test_evaluation": False,
    "release_acceptance_granted": False,
})

print(json.dumps({
    "stage": "EXPAND60_PLAN_FROZEN",
    "queries": 60,
    "sampling": audit60,
    "maximum_seconds": 7200,
    "maximum_new_pairs": 24000,
    "api_calls": 0,
    "plan": LAB["plan"],
}, ensure_ascii=False, indent=2), flush=True)

# Only the reference and the frozen selected method are executed.
run_route({"name": "baseline_D_60", "mode": "baseline"})
run_route({
    "name": "onnx400_fusion_60",
    "mode": "fusion",
    "depth": 400,
})

base60 = LAB["results"]["baseline_D_60"]
after60 = LAB["results"]["onnx400_fusion_60"]
complete60 = base60["status"] == after60["status"] == "COMPLETE"

report60 = {
    "version": "NB22E_EXPAND60_V1",
    "complete": complete60,
    "plan": LAB["plan"],
    "sampling": audit60,
    "routes": {
        name: {k: v for k, v in r.items() if k != "values"}
        for name, r in LAB["results"].items()
    },
    "new_pairs": LAB["new_pairs"],
    "runtime_check_pairs": LAB["parity_pairs"],
    "api_calls": 0,
    "downloads": False,
    "test_evaluation": False,
    "release_acceptance_granted": False,
}

if complete60:
    before60, measured60 = base60["metrics"], after60["metrics"]
    report60["before"] = before60
    report60["after"] = measured60
    report60["delta"] = {
        k: measured60[k] - before60[k] for k in before60
    }
    report60["passes_point_guard"] = eligible(measured60, before60)
    report60["source_slices"] = {}

    for source60 in ("book", "transcript", "website"):
        qs60 = [
            q for q in selected60
            if set(LAB["source_types"][q]) == {source60}
        ]
        b60 = mean_metric(base60["values"][q] for q in qs60)
        a60 = mean_metric(after60["values"][q] for q in qs60)
        report60["source_slices"][source60] = {
            "queries": len(qs60),
            "before": b60,
            "after": a60,
            "delta": {k: a60[k] - b60[k] for k in b60},
        }

    changes60 = [{
        "candidate_id": q,
        "source_types": LAB["source_types"][q],
        "source_component": component60[q],
        "before": base60["values"][q],
        "after": after60["values"][q],
        "delta": {
            k: after60["values"][q][k] - base60["values"][q][k]
            for k in ("recall_at_10", "ndcg_at_10", "hit_at_10")
        },
    } for q in selected60]

    report60["paired_query_changes"] = save(
        "C010.paired_query_changes", changes60, rows=True
    )
    report60["worsened_queries"] = {
        k: sum(r["delta"][k] < -1e-12 for r in changes60)
        for k in ("recall_at_10", "ndcg_at_10", "hit_at_10")
    }

    print("\nارزیابی ۶۰ سؤال جدید Development؛ پذیرش نهایی محصول نیست.")
    print("| معیار | قبل | بعد | تغییر؛ واحد درصد | حد پذیرش | قبول |")
    print("|---|---:|---:|---:|---:|:---:|")

    for key, title in [
        ("recall_at_10", "Recall@10"),
        ("ndcg_at_10", "nDCG@10"),
        ("hit_at_10", "Hit@10"),
    ]:
        limit = CONFIG["thresholds"].get(key)
        threshold = (
            f"{100*limit:.2f}%" if limit is not None else "تعیین نشده"
        )
        mark = (
            "✅" if limit is not None and measured60[key] >= limit else ""
        )
        print(
            f"| {title} | {100*before60[key]:.2f}% | "
            f"{100*measured60[key]:.2f}% | "
            f"{100*(measured60[key]-before60[key]):+.2f} | "
            f"{threshold} | {mark} |"
        )

    report60["next_step"] = (
        "REVIEW_SOURCE_SLICES_AND_REGRESSIONS_BEFORE_LARGER_RUN"
    )
else:
    report60["next_step"] = (
        "RERUN_THIS_SAME_CELL_TO_RESUME_CACHED_WORK_IF_TIME_OR_PAIR_CAP;"
        "_OTHERWISE_REVIEW_HELD_REASON"
    )
    print(
        "ارزیابی هنوز کامل نیست؛ نتیجهٔ جزئی به کل ۶۰ سؤال تعمیم داده نمی‌شود."
    )

receipt60 = save("C010.expand60_report", report60)

print(json.dumps({
    "complete": complete60,
    "passes_point_guard": report60.get("passes_point_guard"),
    "worsened_queries": report60.get("worsened_queries"),
    "source_comparison": {
        source: {
            side: {
                k: values[side][k]
                for k in ("recall_at_10", "ndcg_at_10", "hit_at_10")
            }
            for side in ("before", "after", "delta")
        }
        for source, values in report60.get("source_slices", {}).items()
    },
    "new_pairs": LAB["new_pairs"],
    "runtime_check_pairs": LAB["parity_pairs"],
    "report": receipt60,
    "api_calls": 0,
    "next_step": report60["next_step"],
}, ensure_ascii=False, indent=2))
```

### NB22E-C011 — اقدام ۴: تشخیص افت و آزمایش اصلاح‌ها با کش موجود

**وضعیت قبلی:** ارزیابی ۶۰سؤالی کامل شد، اما روش `onnx400_fusion` به‌دلیل افت Recall و Hit شرط عدم‌افت را پاس نکرد. مرجع NB22D حفظ شده است.

**هدف:** مشخص‌کردن محل ازدست‌رفتن شواهد و بررسی اصلاح‌های محدود، بدون inference تازه.

**این اقدام پنج زیرکار دارد:**
1. بازخوانی و تطبیق رتبه‌ها و سنجه‌های ذخیره‌شدهٔ C010.
2. تفکیک شواهد ازدست‌رفته بین حذف از Pool400، رتبه‌بندی مدل و ترکیب رتبه‌ها.
3. بررسی مدل۴۰۰ به‌تنهایی و دو اصلاح مشخص:
   - ترکیب رتبهٔ مرجع با رتبهٔ مدل، با وزن بیشتر برای مرجع؛
   - حفظ پنج رتبهٔ اول مرجع و تکمیل ادامه با مدل۴۰۰.
4. جدول قبل/بعد، تغییر، حد پذیرش و تفکیک کتاب، صوت و وب.
5. ذخیرهٔ تشخیص سؤال‌به‌سؤال، رتبه‌های آزمایشی و گام بعدی.

**معیارها:** Recall@10≥۸۵٪ و nDCG@10≥۷۵٪؛ Hit@10 حد مطلق مصوب ندارد. عدم‌افت هر نوع منبع نیز جدا بررسی می‌شود.

**API، دانلود و اجرای مدل تازه:** صفر. نبود امتیاز لازم، توقف روشن می‌دهد؛ مدل خودکار اجرا نمی‌شود.

**مرز نتیجه:** همین ۶۰ سؤال قبلاً دیده شده‌اند؛ اصلاح‌ها اکتشافی‌اند و شاهد پذیرش مستقل نیستند. ✅ فقط عبور عددی از حد در همین نمونه را نشان می‌دهد.

**شرط پایان:** بازتولید دقیق نتیجهٔ C010، ثبت محل افت و مقایسهٔ کامل حالت‌های مشخص‌شده. هیچ روش تازه‌ای خودکار جایگزین NB22D نمی‌شود.

```python
# NB22E-C011 — Action 4: offline diagnosis and bounded rank-only experiments.
def audit_cached60():
    require(len(LAB.get("pilot", [])) == 60,
            "C010_COMPLETED_COHORT_REQUIRED")
    require(
        LAB["model_contract"]["specification"].get("schema_version")
        == "NB22E_COHERENT_LOCAL_SCORE_V1",
        "COHERENT_SCORE_CONTRACT_REQUIRED",
    )

    prior = LAB["results"].get("baseline_D_60", {})
    failed = LAB["results"].get("onnx400_fusion_60", {})
    require(
        prior.get("status") == failed.get("status") == "COMPLETE",
        "BOTH_C010_ROUTES_MUST_BE_COMPLETE",
    )
    counts_before = (
        LAB["new_pairs"], LAB["parity_pairs"],
        STATE["api_calls"], STATE["model_pairs"],
    )
    qs = list(LAB["pilot"])
    baseline = read_json(prior["artifact"])
    old_fusion = read_json(failed["artifact"])

    require(
        set(baseline["orders"]) == set(old_fusion["orders"]) == set(qs),
        "SAVED_ROUTE_COHORT_CONFLICT",
    )
    require(
        baseline["complete"] and old_fusion["complete"],
        "SAVED_ROUTES_INCOMPLETE",
    )
    require(
        CONFIG["rrf_k"] == 10 and CONFIG["reranker_weight"] == .75,
        "C010_REPLAY_PARAMETERS_CHANGED",
    )

    plan = save("C011.offline_plan", {
        "version": "NB22E_OFFLINE_DIAGNOSIS_V1",
        "source_plan": LAB["plan"],
        "baseline": prior["artifact"],
        "failed_method": failed["artifact"],
        "cohort": qs,
        "model_contract": LAB["model_contract"],
        "diagnosis": [
            "candidate_pool_exclusion",
            "pure_model_order",
            "fusion_order",
        ],
        "fixed_exploratory_methods": [
            "pure_model400_diagnostic",
            "rrf_reference75_model25_union",
            "preserve_reference_top5_then_model400",
        ],
        "hypotheses": {
            "rrf_reference75_model25_union": (
                "Keep reference candidates eligible through rank fusion; "
                "do not mix raw model scores."
            ),
            "preserve_reference_top5_then_model400": (
                "Preserve reference first five ranks; "
                "fill remaining ranks from model400 without duplicates."
            ),
        },
        "scope": "EXPOSED60_EXPLORATION_NOT_INDEPENDENT_VALIDATION",
        "automatic_replacement": False,
        "new_inference_pairs": 0,
        "api_calls": 0,
        "downloads": False,
        "test_evaluation": False,
    })

    names = (
        "baseline_D_60",
        "failed_fusion400",
        "pure_model400_diagnostic",
        "rrf_reference75_model25_union",
        "preserve_reference_top5_then_model400",
    )
    orders = {name: {} for name in names}
    diagnoses = []
    gold = LAB["dev"]["gold"]
    component_map = components(LAB["dev"])

    for q in qs:
        base = baseline["orders"][q]
        stored_fusion = old_fusion["orders"][q]
        pool = LAB["pools"][q][:400]

        # Read/validate only. Never call score_order or a predictor.
        _, record = score_cache(q)
        require(
            set(pool).issubset(record["pairs"]),
            "MISSING_CACHED_PAIRS_NO_INFERENCE:" + q,
        )
        scores = {
            c: float(record["pairs"][c]["score"]) for c in pool
        }
        pure = sorted(pool, key=lambda c: (-scores[c], c))
        rr = {c: i + 1 for i, c in enumerate(pure)}
        pr = {c: i + 1 for i, c in enumerate(pool)}
        replay = sorted(
            pool,
            key=lambda c: (
                -(.75 / (10 + rr[c]) + .25 / (10 + pr[c])), c
            ),
        )
        require(
            replay == stored_fusion,
            "C010_FUSION_ORDER_REPLAY_CONFLICT:" + q,
        )

        for label, order, expected in (
            ("baseline", base, baseline["metrics_by_query"][q]),
            ("fusion", replay, old_fusion["metrics_by_query"][q]),
        ):
            require(
                all(
                    abs(v - expected[k]) <= 1e-12
                    for k, v in metrics(order, gold[q]).items()
                ),
                "SAVED_METRIC_REPLAY_CONFLICT:" + label + ":" + q,
            )

        # Experiment 1: rank fusion with the original reference ranking.
        br = {c: i + 1 for i, c in enumerate(base)}
        union = set(base) | set(pure)
        protected_rrf = sorted(
            union,
            key=lambda c: (
                -(
                    (.75 / (10 + br[c]) if c in br else 0.0)
                    + (.25 / (10 + rr[c]) if c in rr else 0.0)
                ),
                c,
            ),
        )

        # Experiment 2: preserve reference top5, then use model400.
        prefix = base[:5]
        protected5 = prefix + [
            c for c in pure if c not in set(prefix)
        ]
        protected5 += [
            c for c in base if c not in set(protected5)
        ]

        variants = {
            "baseline_D_60": base,
            "failed_fusion400": replay,
            "pure_model400_diagnostic": pure,
            "rrf_reference75_model25_union": protected_rrf,
            "preserve_reference_top5_then_model400": protected5,
        }
        for name, order in variants.items():
            require(
                len(order) == len(set(order))
                and set(order).issubset(LAB["dev"]["chunks"]),
                "DERIVED_RANK_INTEGRITY_CONFLICT:" + name,
            )
            orders[name][q] = order

        b10, p10, f10 = (
            set(order[:10]) & gold[q]
            for order in (base, pure, replay)
        )
        lost = b10 - f10
        missing_pool = lost - set(pool)
        pushed_model = (lost & set(pool)) - p10
        pushed_fusion = lost & p10

        require(
            missing_pool | pushed_model | pushed_fusion == lost
            and not (
                missing_pool & pushed_model
                or missing_pool & pushed_fusion
                or pushed_model & pushed_fusion
            ),
            "LOST_GOLD_STAGE_PARTITION_CONFLICT",
        )

        base_metrics = metrics(base, gold[q])
        fusion_metrics = metrics(replay, gold[q])
        final_positions = {
            cid: i + 1 for i, cid in enumerate(replay)
        }
        diagnoses.append({
            "candidate_id": q,
            "source_types": LAB["source_types"][q],
            "source_component": component_map[q],
            "gold_count": len(gold[q]),
            "failed_method_delta": {
                k: fusion_metrics[k] - base_metrics[k]
                for k in (
                    "recall_at_10", "ndcg_at_10", "hit_at_10"
                )
            },
            "baseline_top10_gold_lost": sorted(lost),
            "lost_gold_outside_pool400": sorted(missing_pool),
            "lost_gold_pushed_down_by_model": sorted(pushed_model),
            "lost_gold_pushed_down_by_fusion": sorted(pushed_fusion),
            "new_gold_gained_in_final10": sorted(f10 - b10),
            "pool400_top10_recall_ceiling": (
                min(10, len(set(pool) & gold[q])) / len(gold[q])
            ),
            "gold_rank_trace": [{
                "chunk_id": c,
                "baseline_rank": br.get(c),
                "pool_rank": pr.get(c),
                "model_rank": rr.get(c),
                "final_rank": final_positions.get(c),
            } for c in sorted(gold[q])],
        })

    per_query = {
        name: {q: metrics(order[q], gold[q]) for q in qs}
        for name, order in orders.items()
    }
    reference = mean_metric(
        per_query["baseline_D_60"].values()
    )
    comparisons = []

    for name, values in per_query.items():
        aggregate = mean_metric(values.values())
        slices = {}

        for source in ("book", "transcript", "website"):
            ids = [
                q for q in qs
                if set(LAB["source_types"][q]) == {source}
            ]
            b = mean_metric(
                per_query["baseline_D_60"][q] for q in ids
            )
            a = mean_metric(values[q] for q in ids)
            slices[source] = {
                "queries": len(ids),
                "before": b,
                "after": a,
                "delta": {k: a[k] - b[k] for k in b},
            }

        global_guard = eligible(aggregate, reference)
        slice_guard = all(
            s["after"][k] >= s["before"][k] - 1e-12
            for s in slices.values()
            for k in ("recall_at_10", "ndcg_at_10", "hit_at_10")
        )
        comparisons.append({
            "method": name,
            "before": reference,
            "after": aggregate,
            "delta": {
                k: aggregate[k] - reference[k] for k in reference
            },
            "aggregate_point_guard": global_guard,
            "source_point_no_regression": slice_guard,
            "source_slices": slices,
            "scope": "EXPLORATORY_ALREADY_EXPOSED60_NOT_VALIDATION",
            "release_acceptance_granted": False,
        })

    candidates = [
        r for r in comparisons
        if r["method"] in (
            "rrf_reference75_model25_union",
            "preserve_reference_top5_then_model400",
        )
        and r["aggregate_point_guard"]
        and r["source_point_no_regression"]
    ]
    shortlist = sorted(
        candidates,
        key=lambda r: (-r["after"]["ndcg_at_10"], r["method"]),
    )

    diagnosis_receipt = save(
        "C011.query_diagnosis", diagnoses, rows=True
    )
    ranking_receipt = save(
        "C011.cached_counterfactual_rankings",
        [{
            "candidate_id": q,
            "method": name,
            "split": "development",
            "ranked_chunk_ids": order[q],
            "metrics": per_query[name][q],
        } for name, order in orders.items() for q in qs],
        rows=True,
    )

    counts_after = (
        LAB["new_pairs"], LAB["parity_pairs"],
        STATE["api_calls"], STATE["model_pairs"],
    )
    require(
        counts_before == counts_after,
        "OFFLINE_AUDIT_CHANGED_EXECUTION_COUNTERS",
    )

    cause_summary = {}
    stage_fields = (
        "lost_gold_outside_pool400",
        "lost_gold_pushed_down_by_model",
        "lost_gold_pushed_down_by_fusion",
    )
    for source in ("book", "transcript", "website"):
        source_rows = [
            r for r in diagnoses
            if set(r["source_types"]) == {source}
        ]
        cause_summary[source] = {
            "queries": len(source_rows),
            "recall_regression_queries": sum(
                r["failed_method_delta"]["recall_at_10"] < -1e-12
                for r in source_rows
            ),
            "lost_gold_units_by_stage": {
                field: sum(len(r[field]) for r in source_rows)
                for field in stage_fields
            },
            "mean_pool400_recall_ceiling": statistics.fmean(
                r["pool400_top10_recall_ceiling"]
                for r in source_rows
            ),
        }

    report = {
        "version": "NB22E_OFFLINE_DIAGNOSIS_V1",
        "status": "COMPLETE_IN_SCOPE",
        "plan": plan,
        "causes": cause_summary,
        "comparisons": comparisons,
        "exploratory_shortlist": [
            r["method"] for r in shortlist
        ],
        "diagnosis": diagnosis_receipt,
        "rankings": ranking_receipt,
        "scope": "POSTHOC_ALREADY_EXPOSED_DEVELOPMENT_NO_NEW_VALIDATION",
        "new_model_pairs": 0,
        "new_api_calls": 0,
        "automatic_replacement": False,
        "release_acceptance_granted": False,
        "next_step": (
            "REVIEW_CASES_AND_FREEZE_ONE_METHOD_BEFORE_SEPARATE_EVALUATION"
            if shortlist else
            "KEEP_NB22D_AND_USE_STAGE_DIAGNOSIS_TO_DEFINE_DIFFERENT_METHOD"
        ),
    }
    receipt = save("C011.offline_report", report)

    print(
        "\nمقایسهٔ آزمایشی روی همان ۶۰ سؤال دیده‌شده؛ "
        "پذیرش مستقل نیست."
    )
    print(
        "| روش | معیار | قبل | بعد | تغییر؛ واحد درصد | "
        "حد پذیرش | قبول |"
    )
    print("|---|---|---:|---:|---:|---:|:---:|")

    for row in comparisons:
        for key, title in (
            ("recall_at_10", "Recall@10"),
            ("ndcg_at_10", "nDCG@10"),
            ("hit_at_10", "Hit@10"),
        ):
            limit = CONFIG["thresholds"].get(key)
            threshold = (
                f"{100*limit:.2f}%"
                if limit is not None else "تعیین نشده"
            )
            mark = (
                "✅"
                if limit is not None and row["after"][key] >= limit
                else ""
            )
            print(
                f"| {row['method']} | {title} | "
                f"{100*row['before'][key]:.2f}% | "
                f"{100*row['after'][key]:.2f}% | "
                f"{100*row['delta'][key]:+.2f} | "
                f"{threshold} | {mark} |"
            )

    source_changes = {
        r["method"]: {
            source: {
                k: values["delta"][k]
                for k in (
                    "recall_at_10", "ndcg_at_10", "hit_at_10"
                )
            }
            for source, values in r["source_slices"].items()
        }
        for r in comparisons
    }
    print(json.dumps({
        "complete": True,
        "causes": cause_summary,
        "source_deltas": source_changes,
        "exploratory_shortlist": report["exploratory_shortlist"],
        "report": receipt,
        "new_model_pairs": 0,
        "new_api_calls": 0,
        "automatic_replacement": False,
        "next_step": report["next_step"],
    }, ensure_ascii=False, indent=2))
    return report


try:
    C011 = audit_cached60()
except InputGap as error:
    C011 = {
        "status": "BLOCKED_INPUT_OR_CONTRACT",
        "reason": str(error),
        "new_model_pairs": 0,
        "new_api_calls": 0,
    }
    print(json.dumps(C011, ensure_ascii=False, indent=2))
```

### NB22E-C012 — اقدام ۵: بررسی محل شاهد در پنجرهٔ ورودی مدل

**شاهد قبلی:** چهار شاهدِ موجود در Pool400، در رتبه‌بندی مدل از Top10 خارج شده‌اند؛ سه مورد صوت و یک مورد وب.

**هدف:** مشخص‌کردن اینکه بازهٔ ثبت‌شدهٔ شاهد در ورودی ۵۱۲توکنی مدل باقی مانده، بخشی از آن حذف شده، کاملاً بیرون افتاده یا اتصال دقیق آن هنوز احراز نشده است.

**زیرکارها:**
1. مصرف چهار مورد از گزارش C011.
2. بازسازی ورودی با tokenizer محلیِ همان Revision.
3. اتصال بازهٔ ثبت‌شدهٔ Golden به سند اصلی و متن قطعه، در صورت تطبیق دقیق.
4. ثبت تعداد توکن، محدودهٔ باقی‌مانده و وضعیت هر شاهد.
5. ذخیره و بازخوانی گزارش، بدون تغییر رتبه، امتیاز، Gold یا معیار پذیرش.

**محدودیت:** وجود یا حذف یک بازهٔ متنی، علت معنایی امتیازدهی مدل را به‌تنهایی اثبات نمی‌کند. اگر اتصال دقیق متن احراز نشود، نتیجه نامشخص ثبت می‌شود.

**API، دانلود و inference مدل:** صفر؛ فقط پردازش محلی tokenizer.

**شرط پایان:** وضعیت هر چهار مورد با شاهد فنی یا دلیل دقیقِ عدم احراز ثبت شود.

</div>

```python
# NB22E-C012 — Local tokenizer/evidence-position audit; no model inference.
def audit_token_window():
    require(
        isinstance(C011, dict)
        and C011.get("status") == "COMPLETE_IN_SCOPE",
        "C011_COMPLETED_DIAGNOSIS_REQUIRED",
    )
    before_counts = (
        LAB["new_pairs"], LAB["parity_pairs"],
        STATE["api_calls"], STATE["model_pairs"],
    )
    diagnoses = read_rows(C011["diagnosis"])
    cases = [
        (r, cid)
        for r in diagnoses
        for cid in r["baseline_top10_gold_lost"]
    ]
    require(bool(cases), "NO_LOST_GOLD_UNITS_TO_AUDIT")

    verified = read_json(PINS["C056"])
    golden = keyed(
        read_rows(verified["input_file_receipts"]["golden"]),
        "candidate_id",
    )

    from transformers import AutoTokenizer

    root = Path(
        os.environ.get("HF_HUB_CACHE")
        or str(
            Path(os.environ.get(
                "HF_HOME", str(Path.home() / ".cache/huggingface")
            )) / "hub"
        )
    )
    snapshot = (
        root
        / ("models--" + MODEL["model_id"].replace("/", "--"))
        / "snapshots"
        / MODEL["revision"]
    )
    require(
        snapshot.is_dir(),
        "PINNED_TOKENIZER_NOT_CACHED_NO_DOWNLOAD",
    )
    tokenizer = AutoTokenizer.from_pretrained(
        str(snapshot),
        local_files_only=True,
        trust_remote_code=False,
    )
    require(
        tokenizer.is_fast,
        "FAST_TOKENIZER_OFFSETS_REQUIRED_NO_DOWNLOAD",
    )

    # Binding declared by the reviewed NB18 context manifest.
    document_binding = {
        "path": (
            "data/phase2/processed/"
            "documents_dedup_quality_accepted_v1.jsonl"
        ),
        "file_bytes_sha256": (
            "9d72c709f81b1218c5aac064052c7ed1"
            "da34869f671e6e32762864e0ad70ea40"
        ),
    }
    target_documents = {
        m["document_id"]
        for _, cid in cases
        for m in LAB["dev"]["chunks"][cid]["member_locators"]
    }
    documents = {}
    document_issue = None

    try:
        p = path_for(document_binding["path"])
        require(
            p.is_file(),
            "BOUND_CANONICAL_DOCUMENTS_NOT_AVAILABLE",
        )
        stamp = p.stat()
        h = hashlib.sha256()

        with p.open("rb") as f:
            for raw in f:
                h.update(raw)
                row = decode(raw)
                if row.get("document_id") in target_documents:
                    require(
                        row["document_id"] not in documents,
                        "DUPLICATE_CANONICAL_DOCUMENT_ID",
                    )
                    documents[row["document_id"]] = row

        after = p.stat()
        require(
            (stamp.st_size, stamp.st_mtime_ns, stamp.st_ino)
            == (after.st_size, after.st_mtime_ns, after.st_ino)
            and h.hexdigest()
            == document_binding["file_bytes_sha256"],
            "CANONICAL_DOCUMENT_HASH_CONFLICT",
        )
    except InputGap as error:
        documents = {}
        document_issue = str(error)

    results = []

    for row, cid in cases:
        q = row["candidate_id"]
        chunk = LAB["dev"]["chunks"][cid]
        text = chunk["content"]
        question = LAB["questions"][q]

        full = tokenizer(
            question, text,
            truncation=False,
            add_special_tokens=True,
        )
        encoded = tokenizer(
            question, text,
            truncation=True,
            max_length=512,
            padding=True,
            return_offsets_mapping=True,
            add_special_tokens=True,
        )
        sequence = encoded.sequence_ids()
        ranges = [
            (a, b)
            for sid, (a, b) in zip(
                sequence, encoded["offset_mapping"]
            )
            if sid == 1 and b > a
        ]
        lo = min((a for a, b in ranges), default=None)
        hi = max((b for a, b in ranges), default=None)
        evidence = []

        for reference in golden[q]["evidence_locators"]:
            key = canonical(reference["source_locator"])

            for member in chunk["member_locators"]:
                if canonical(member["locator"]) != key:
                    continue

                document = documents.get(
                    member["document_id"], {}
                )
                content = document.get("content")
                a = reference.get("text_char_start")
                b = reference.get("text_char_end")
                item = {
                    "document_id": member["document_id"],
                    "source_char_start": a,
                    "source_char_end": b,
                    "status": (
                        "UNRESOLVED_SOURCE_SPAN_OR_TEXT_BINDING"
                    ),
                }

                if (
                    isinstance(content, str)
                    and type(a) is int and type(b) is int
                    and 0 <= a < b <= len(content)
                ):
                    quote = content[a:b].strip()
                    positions = []
                    start = 0

                    while quote:
                        pos = text.find(quote, start)
                        if pos < 0:
                            break
                        positions.append(
                            (pos, pos + len(quote))
                        )
                        start = pos + 1

                    if positions:
                        fully_kept = any(
                            lo is not None
                            and lo <= a0 and b0 <= hi
                            for a0, b0 in positions
                        )
                        partly_kept = any(
                            lo is not None
                            and a0 < hi and b0 > lo
                            for a0, b0 in positions
                        )
                        status = (
                            "REGISTERED_SPAN_RETAINED"
                            if fully_kept else
                            "REGISTERED_SPAN_PARTLY_TRUNCATED"
                            if partly_kept else
                            "REGISTERED_SPAN_OUTSIDE_WINDOW"
                        )
                        item.update(
                            status=status,
                            exact_chunk_occurrences=[
                                list(x) for x in positions
                            ],
                            registered_span_text=quote,
                        )
                    else:
                        item["status"] = (
                            "REGISTERED_SOURCE_SLICE_NOT_"
                            "EXACTLY_MAPPED_TO_CHUNK"
                        )

                evidence.append(item)

        statuses = [x["status"] for x in evidence]
        results.append({
            "candidate_id": q,
            "chunk_id": cid,
            "source_types": row["source_types"],
            "question": question,
            "chunk_content_sha256": digest(text.encode()),
            "pair_tokens_without_truncation": len(
                full["input_ids"]
            ),
            "actual_pair_tokens": len(encoded["input_ids"]),
            "full_document_tokens": len(
                tokenizer(
                    text, add_special_tokens=False
                )["input_ids"]
            ),
            "retained_document_tokens": len(ranges),
            "retained_chunk_char_range": [lo, hi],
            "pair_exceeds_512": len(full["input_ids"]) > 512,
            "truncation_side": tokenizer.truncation_side,
            "registered_evidence_statuses": statuses or [
                "UNRESOLVED_NO_REFERENCE_MEMBER_MATCH"
            ],
            "registered_evidence": evidence,
            "retained_text_preview": (
                text[lo:min(hi, lo + 600)]
                if lo is not None else ""
            ),
            "text_immediately_after_window": (
                text[hi:hi + 600]
                if hi is not None else ""
            ),
            "semantic_reason_for_bad_ranking_proven": False,
        })

    require(
        before_counts == (
            LAB["new_pairs"], LAB["parity_pairs"],
            STATE["api_calls"], STATE["model_pairs"],
        ),
        "TOKENIZER_AUDIT_CHANGED_INFERENCE_COUNTERS",
    )

    receipt = save(
        "C012.token_window_cases", results, rows=True
    )
    counts = dict(collections.Counter(
        status
        for r in results
        for status in r["registered_evidence_statuses"]
    ))
    summary = {
        "version": "NB22E_TOKEN_WINDOW_AUDIT_V1",
        "status": "COMPLETE_IN_SCOPE",
        "lost_query_chunk_pairs": len(results),
        "pairs_exceeding_512": sum(
            r["pair_exceeds_512"] for r in results
        ),
        "registered_evidence_status_counts": counts,
        "canonical_document_issue": document_issue,
        "cases": receipt,
        "diagnosis_source": C011["diagnosis"],
        "document_binding": document_binding,
        "tokenizer": {
            "model_id": MODEL["model_id"],
            "revision": MODEL["revision"],
            "class": type(tokenizer).__name__,
            "fast": True,
        },
        "conclusion_limit": (
            "Registered source span coverage only; "
            "semantic cause and source-span qualification "
            "not independently established."
        ),
        "new_model_pairs": 0,
        "new_api_calls": 0,
        "downloads": False,
        "automatic_replacement": False,
        "release_acceptance_granted": False,
    }
    report = save("C012.token_window_report", summary)

    print(
        "| منبع | سؤال | توکن جفت کامل | "
        "ورودی واقعی | وضعیت بازهٔ شاهد |"
    )
    print("|---|---|---:|---:|---|")

    for r in results:
        print(
            f"| {','.join(r['source_types'])} | "
            f"{r['candidate_id']} | "
            f"{r['pair_tokens_without_truncation']} | "
            f"{r['actual_pair_tokens']} | "
            f"{', '.join(r['registered_evidence_statuses'])} |"
        )

    print(json.dumps(
        {**summary, "report": report},
        ensure_ascii=False,
        indent=2,
    ))
    return summary


try:
    C012 = audit_token_window()
except InputGap as error:
    C012 = {
        "status": "BLOCKED_INPUT_OR_CONTRACT",
        "reason": str(error),
        "new_model_pairs": 0,
        "new_api_calls": 0,
    }
    print(json.dumps(C012, ensure_ascii=False, indent=2))
```

### NB22E-C013 — اقدام ۶: تهیهٔ پروندهٔ مقایسهٔ متن‌ها

**هدف:** بررسی متن چهار شاهدِ دارای افت در برابر قطعه‌های بالاتر در رتبه‌بندی مدل.

**شاهد قبلی:** هر چهار قطعه داخل Pool400 بوده‌اند و ورودی آن‌ها زیر ۵۱۲ توکن است؛ بریده‌شدن متن در این موارد مشاهده نشد.

**زیرکارها:**
1. دریافت سؤال، پاسخ مرجع ثبت‌شده و متن کامل شاهد.
2. دریافت دو قطعهٔ خارج از Gold که مدل بالاتر قرار داده است؛ یکی از بالای فهرست و دیگری نزدیک‌تر به شاهد.
3. ثبت رتبه‌ها، امتیازها، اختلاف امتیاز و نتیجهٔ بررسی توکن.
4. ذخیره و بازخوانی پروندهٔ تجمیعی برای بررسی محتوایی.

**محدودیت:** خارج‌بودن از Gold، نامرتبط‌بودن قطعه را ثابت نمی‌کند. هنوز رأی معنایی یا علت قطعی خطای مدل تعیین نمی‌شود.

**API، دانلود و inference:** صفر.

**شرط پایان:** متن کامل و منشأ هر چهار پرونده در یک فایل JSONL معتبر آماده باشد.

</div>

```python
# NB22E-C013 — Export contrastive cases from existing files/cache only.
def build_contrast_cases():
    require(C011.get("status") == "COMPLETE_IN_SCOPE", "C011_REQUIRED")
    require(C012.get("status") == "COMPLETE_IN_SCOPE", "C012_REQUIRED")

    before = (
        LAB["new_pairs"], LAB["parity_pairs"],
        STATE["api_calls"], STATE["model_pairs"],
    )
    diagnoses = read_rows(C011["diagnosis"])
    token_rows = {
        (r["candidate_id"], r["chunk_id"]): r
        for r in read_rows(C012["cases"])
    }

    pure_orders = {}
    for row in read_rows(C011["rankings"]):
        if row["method"] == "pure_model400_diagnostic":
            require(
                row["candidate_id"] not in pure_orders,
                "DUPLICATE_MODEL_ORDER",
            )
            pure_orders[row["candidate_id"]] = row["ranked_chunk_ids"]

    source = read_json(PINS["C056"])
    golden = keyed(
        read_rows(source["input_file_receipts"]["golden"]),
        "candidate_id",
    )
    cases = []

    for row in diagnoses:
        q = row["candidate_id"]
        if not row["baseline_top10_gold_lost"]:
            continue

        order = pure_orders[q]
        require(len(order) == len(set(order)), "MODEL_ORDER_DUPLICATE")
        positions = {cid: i + 1 for i, cid in enumerate(order)}
        _, cache = score_cache(q)  # Read only; no predictor call.

        def packet(cid):
            chunk = LAB["dev"]["chunks"][cid]
            item = cache["pairs"].get(cid)
            require(
                item is not None,
                "CACHED_SCORE_REQUIRED_NO_INFERENCE:" + cid,
            )
            require(
                item["content_sha256"] == digest(chunk["content"].encode()),
                "CHUNK_TEXT_IDENTITY_CONFLICT",
            )
            return {
                "chunk_id": cid,
                "source_id": chunk.get("source_id"),
                "source_type": chunk.get("source_type"),
                "content": chunk["content"],
                "content_sha256": item["content_sha256"],
                "model_score": item["score"],
                "model_rank": positions.get(cid),
                "in_frozen_gold": cid in LAB["dev"]["gold"][q],
                "member_locators": chunk["member_locators"],
            }

        for cid in row["baseline_top10_gold_lost"]:
            require(
                cid in positions,
                "LOST_GOLD_NOT_IN_MODEL_POOL:" + cid,
            )
            earlier = [
                c for c in order[:positions[cid] - 1]
                if c not in LAB["dev"]["gold"][q]
            ]
            competitor_ids = (
                list(dict.fromkeys([earlier[0], earlier[-1]]))
                if earlier else []
            )

            witness = packet(cid)
            competitors = [packet(c) for c in competitor_ids]
            for competitor in competitors:
                competitor["score_margin_over_reference"] = (
                    competitor["model_score"] - witness["model_score"]
                )

            cases.append({
                "candidate_id": q,
                "source_types": row["source_types"],
                "question": LAB["questions"][q],
                "recorded_expected_answer": (
                    golden[q]["evaluation_content"].get("expected_answer")
                ),
                "recorded_gold_locators": golden[q]["evidence_locators"],
                "rank_trace": next(
                    (
                        r for r in row["gold_rank_trace"]
                        if r["chunk_id"] == cid
                    ),
                    None,
                ),
                "reference_chunk": witness,
                "higher_ranked_not_in_gold": competitors,
                "token_window_audit": token_rows.get((q, cid)),
                "semantic_verdict": "NOT_REVIEWED",
                "warning": (
                    "Not-in-Gold is not proof of irrelevance; "
                    "expected answer and Gold are historical references."
                ),
            })

    expected = sum(
        len(r["baseline_top10_gold_lost"]) for r in diagnoses
    )
    require(
        len(cases) == expected and expected > 0,
        "CONTRAST_CASE_COUNT_CONFLICT",
    )
    require(
        before == (
            LAB["new_pairs"], LAB["parity_pairs"],
            STATE["api_calls"], STATE["model_pairs"],
        ),
        "OFFLINE_EXPORT_CHANGED_EXECUTION_COUNTERS",
    )

    receipt = save("C013.contrast_cases", cases, rows=True)
    summary = {
        "version": "NB22E_CONTRAST_CASES_V1",
        "status": "COMPLETE_IN_SCOPE",
        "case_count": len(cases),
        "cases": receipt,
        "diagnosis": C011["diagnosis"],
        "token_audit": C012["cases"],
        "scope": "CASE_PACK_ONLY_NO_SEMANTIC_VERDICT_OR_ACCEPTANCE",
        "new_model_pairs": 0,
        "new_api_calls": 0,
        "release_acceptance_granted": False,
    }
    report = save("C013.contrast_report", summary)

    print("| منبع | سؤال | رتبهٔ مدلِ شاهد | تعداد قطعهٔ مقایسه |")
    print("|---|---|---:|---:|")
    for case in cases:
        print(
            f"| {','.join(case['source_types'])} | "
            f"{case['candidate_id']} | "
            f"{case['reference_chunk']['model_rank']} | "
            f"{len(case['higher_ranked_not_in_gold'])} |"
        )

    print(json.dumps(
        {**summary, "report": report},
        ensure_ascii=False,
        indent=2,
    ))
    print(
        "\nفایل JSONL مشخص‌شده در cases.path را ارسال کن؛ "
        "متن کامل پرونده‌ها داخل آن است."
    )
    return summary


try:
    C013 = build_contrast_cases()
except InputGap as error:
    C013 = {
        "status": "BLOCKED_INPUT_OR_CONTRACT",
        "reason": str(error),
        "new_model_pairs": 0,
        "new_api_calls": 0,
    }
    print(json.dumps(C013, ensure_ascii=False, indent=2))
```

### NB22E-C014 — اقدام ۷: تعیین امکانات آزمایش مدل جایگزین

**هدف:** شناسایی مدل‌های موجود در کش محلی و توان اجرای دستگاه، برای طراحی یک آزمایش متفاوت با مرجع NB22D.

**زیرکارها:**
1. ثبت RAM، معماری پردازنده و نسخهٔ کتابخانه‌های مرتبط.
2. بررسی Metadata مدل‌های کش‌شده و وجود فایل‌های وزن.
3. جداکردن مدل فعلی، مدل‌های Embedding و گزینه‌های نیازمند بررسی برای رتبه‌بندی.
4. ثبت نتیجه و تعیین پیش‌نیاز آزمایش بعدی.

**محدودیت:** وجود فایل وزن یا معماری Classification، مناسب‌بودن مدل برای Reranking فارسی را ثابت نمی‌کند؛ صلاحیت و قابلیت اجرای گزینهٔ انتخابی جدا بررسی می‌شوند.

**API، دانلود و inference:** صفر. فایل‌های وزن باز یا اجرا نمی‌شوند؛ فقط مشخصات و موجودی خوانده می‌شود.

**شرط پایان:** موجودی قابل بررسی و امکانات دستگاه مشخص باشند. هیچ مدل تازه‌ای خودکار نصب، دانلود یا اجرا نمی‌شود.

</div>

```python
# NB22E-C014 — Offline model/runtime inventory; no inference or downloads.
import importlib.metadata as _metadata
import platform as _platform
import subprocess as _subprocess


def inspect_local_models():
    packages = {}
    for name in (
        "torch", "transformers", "sentence-transformers",
        "onnxruntime", "numpy", "tokenizers",
    ):
        try:
            packages[name] = _metadata.version(name)
        except _metadata.PackageNotFoundError:
            packages[name] = None

    ram = None
    try:
        if _platform.system() == "Darwin":
            ram = int(_subprocess.check_output(
                ["sysctl", "-n", "hw.memsize"],
                text=True,
                timeout=5,
            ))
        else:
            ram = (
                os.sysconf("SC_PAGE_SIZE")
                * os.sysconf("SC_PHYS_PAGES")
            )
    except (OSError, ValueError, _subprocess.SubprocessError):
        pass

    roots = []
    proposed = [
        os.environ.get("HF_HUB_CACHE"),
        os.environ.get("HUGGINGFACE_HUB_CACHE"),
        str(
            Path(os.environ.get(
                "HF_HOME", str(Path.home() / ".cache/huggingface")
            )) / "hub"
        ),
        os.environ.get("TRANSFORMERS_CACHE"),
    ]
    for value in proposed:
        if value:
            p = Path(value).expanduser()
            if p.is_dir() and p.resolve() not in roots:
                roots.append(p.resolve())

    inventory, issues = [], []

    for cache_root in roots:
        for folder in sorted(cache_root.glob("models--*")):
            if not folder.is_dir():
                continue
            model_id = (
                folder.name.removeprefix("models--").replace("--", "/")
            )

            for snapshot in sorted((folder / "snapshots").glob("*")):
                if not snapshot.is_dir():
                    continue

                config_path = snapshot / "config.json"
                config = {}
                config_status = "NOT_AVAILABLE"

                if config_path.is_file():
                    try:
                        config = decode(config_path.read_bytes())
                        require(
                            isinstance(config, dict),
                            "MODEL_CONFIG_OBJECT_REQUIRED",
                        )
                        config_status = "READ_METADATA_ONLY"
                    except (ValueError, UnicodeError, InputGap) as error:
                        issues.append({
                            "model_id": model_id,
                            "revision": snapshot.name,
                            "issue": (
                                "CONFIG_PARSE_FAILED:"
                                + type(error).__name__
                            ),
                        })
                        continue

                singles = [
                    p for p in (
                        snapshot / "model.safetensors",
                        snapshot / "pytorch_model.bin",
                    )
                    if p.is_file()
                ]
                weight_files = list(singles)
                indexes = []

                for filename in (
                    "model.safetensors.index.json",
                    "pytorch_model.bin.index.json",
                ):
                    index_path = snapshot / filename
                    if not index_path.is_file():
                        continue
                    try:
                        data = decode(index_path.read_bytes())
                        names = sorted(set(
                            data.get("weight_map", {}).values()
                        ))
                        require(
                            bool(names)
                            and all(
                                isinstance(n, str)
                                and not Path(n).is_absolute()
                                and ".." not in Path(n).parts
                                for n in names
                            ),
                            "INVALID_WEIGHT_INDEX",
                        )
                        complete = all(
                            (snapshot / n).is_file() for n in names
                        )
                        indexes.append({
                            "index": filename,
                            "shards": len(names),
                            "all_shards_present": complete,
                        })
                        weight_files.extend(
                            snapshot / n for n in names
                            if (snapshot / n).is_file()
                        )
                    except (
                        ValueError, TypeError, AttributeError, InputGap
                    ) as error:
                        indexes.append({
                            "index": filename,
                            "issue": type(error).__name__,
                        })

                unique_weights = {
                    str(p.resolve()): p for p in weight_files
                }
                onnx = [
                    p for p in (snapshot / "onnx").glob("*.onnx")
                    if p.is_file()
                ]
                if (snapshot / "model.onnx").is_file():
                    onnx.append(snapshot / "model.onnx")

                architectures = config.get("architectures") or []
                if not isinstance(architectures, list):
                    architectures = []

                classifier = any(
                    "ForSequenceClassification" in str(a)
                    or "CrossEncoder" in str(a)
                    for a in architectures
                )
                same_id = model_id == MODEL["model_id"]
                same_snapshot = (
                    same_id and snapshot.name == MODEL["revision"]
                )
                full_weights = bool(singles) or any(
                    i.get("all_shards_present") for i in indexes
                )
                tokenizer_files = [
                    n for n in (
                        "tokenizer.json", "tokenizer_config.json",
                        "vocab.txt", "vocab.json", "merges.txt",
                        "sentencepiece.bpe.model", "tokenizer.model",
                    )
                    if (snapshot / n).is_file()
                ]

                inventory.append({
                    "model_id": model_id,
                    "revision": snapshot.name,
                    "is_current_model": same_id,
                    "is_current_snapshot": same_snapshot,
                    "configuration": config_status,
                    "architectures": architectures,
                    "model_type": config.get("model_type"),
                    "hidden_size": config.get("hidden_size"),
                    "num_layers": config.get("num_hidden_layers"),
                    "torch_weight_set_present": full_weights,
                    "torch_weights_gib": round(
                        sum(
                            p.stat().st_size
                            for p in unique_weights.values()
                        ) / 1024**3,
                        3,
                    ),
                    "weight_indexes": indexes,
                    "onnx_files": [
                        p.relative_to(snapshot).as_posix()
                        for p in onnx
                    ],
                    "tokenizer_files": tokenizer_files,
                    "declares_custom_auto_map": bool(
                        config.get("auto_map")
                    ),
                    "classification_architecture": classifier,
                    "alternative_for_qualification": (
                        not same_id
                        and classifier
                        and (full_weights or bool(onnx))
                        and bool(tokenizer_files)
                    ),
                    "reranker_task_and_persian_support_verified": False,
                    "weight_integrity_and_runtime_verified": False,
                })

    alternatives = [
        r for r in inventory if r["alternative_for_qualification"]
    ]
    summary = {
        "version": "NB22E_LOCAL_INVENTORY_V1",
        "status": "COMPLETE_IN_SCOPE",
        "environment": {
            "python": _platform.python_version(),
            "system": _platform.system(),
            "machine": _platform.machine(),
            "logical_cpus": os.cpu_count(),
            "ram_gib": (
                round(ram / 1024**3, 1)
                if ram is not None else None
            ),
            "packages": packages,
        },
        "cache_roots_checked": len(roots),
        "models": inventory,
        "issues": issues,
        "alternative_snapshots_for_qualification": len(alternatives),
        "scope": (
            "LOCAL_CACHE_METADATA_ONLY;"
            "_NO_VERIFIED_REPLACEMENT_MODEL_YET"
        ),
        "legacy_transformers_cache_layout_not_exhaustively_scanned": True,
        "new_model_pairs": 0,
        "new_api_calls": 0,
        "downloads": False,
        "next_step": (
            "QUALIFY_CACHED_ALTERNATIVE_THEN_FREEZE_BOUNDED_COMPARISON"
            if alternatives else
            "NO_READY_ALTERNATIVE_FOUND_IN_CHECKED_CACHES;"
            "_PLAN_ONE_SEPARATE_MODEL_DOWNLOAD_BEFORE_ANY_EXECUTION"
        ),
    }
    receipt = save("C014.local_model_inventory", summary)

    print(
        "| مدل | Revision | وزن Torch کامل | ONNX | "
        "گزینهٔ نیازمند بررسی |"
    )
    print("|---|---|:---:|---:|:---:|")
    for r in inventory:
        print(
            f"| {r['model_id']} | {r['revision'][:12]} | "
            f"{'موجود' if r['torch_weight_set_present'] else 'احراز نشد'} | "
            f"{len(r['onnx_files'])} | "
            f"{'بله' if r['alternative_for_qualification'] else '—'} |"
        )

    print(json.dumps(
        {**summary, "report": receipt},
        ensure_ascii=False,
        indent=2,
    ))
    return summary


C014 = inspect_local_models()
```

### NB22E-C015 — اقدام ۸: دریافت نسخهٔ مشخص مدل جایگزین

**هدف:** آماده‌کردن مدل متفاوت BGE برای آزمایش بعدی، بدون اجرای مدل.

**زیرکارها:**
1. ثبت مدل، Revision، فایل‌های لازم و حجم دانلود.
2. دریافت فقط وزن Safetensors و فایل‌های tokenizer/config.
3. کنترل حجم و هش فایل‌ها در برابر اطلاعات منبع اصلی.
4. ذخیرهٔ رسید نسخهٔ آماده؛ بدون تغییر مدل و کش قبلی.

**شبکه:** دانلود حدود ۲٫۳ گیگابایت از Hugging Face؛ مصرف حجم اینترنت دارد.

**API inference، نصب کتابخانه و اجرای مدل:** صفر.

**شرط پایان:** تمام فایل‌های نسخهٔ مشخص موجود و هویت آن‌ها تأیید شود. این مرحله، کیفیت بهتر یا سرعت مناسب مدل را اثبات نمی‌کند.

**مرحلهٔ بعد:** آزمایش کوچک زمان و حافظه روی CPU، پیش از ارزیابی کیفیت یا اجرای بزرگ.

</div>

```python
# C015 v2 — Local model verification only
# بدون دانلود، بدون API و بدون اجرای مدل

import os
import json
import hashlib
from pathlib import Path

os.environ["TOKENIZERS_PARALLELISM"] = "false"

BGE_MODEL_ID = "BAAI/bge-reranker-v2-m3"
BGE_REVISION = "953dc6f6f85a1b2dbfca4c34a2796e7dde08d41e"

# همان مسیر صریحی که دستور دانلود ترمینال استفاده می‌کند
BGE_LOCAL_PATH = (
    Path.home()
    / ".cache/huggingface/hub"
    / "models--BAAI--bge-reranker-v2-m3"
    / "snapshots"
    / BGE_REVISION
)

# نوع هش فایل‌های کوچک: Git blob SHA1
# نوع هش فایل‌های بزرگ: SHA256
BGE_EXPECTED_FILES = {
    "config.json": (
        795, "git_sha1",
        "9f62673cb00ec41dcec8947b9ed16f6f2eb23ba2",
    ),
    "model.safetensors": (
        2271071852, "sha256",
        "d9e3e081faff1eefb84019509b2f5558fd74c1a05a2c7db22f74174fcedb5286",
    ),
    "sentencepiece.bpe.model": (
        5069051, "sha256",
        "cfc8146abe2a0488e9e2a0c56de7952f7c11ab059eca145a0a727afce0db2865",
    ),
    "special_tokens_map.json": (
        964, "git_sha1",
        "b1879d702821e753ffe4245048eee415d54a9385",
    ),
    "tokenizer.json": (
        17098273, "sha256",
        "69564b696052886ed0ac63fa393e928384e0f8caada38c1f4864a9bfbf379c15",
    ),
    "tokenizer_config.json": (
        1173, "git_sha1",
        "328a00a9a560aadcf2a3064f917517359eb3cc26",
    ),
}

# در صورت شکست، نتیجهٔ موفق اجرای قبلی باقی نماند
C015 = {
    "version": "NB22E_LOCAL_MODEL_VERIFICATION_V2",
    "status": "VERIFYING",
    "model_id": BGE_MODEL_ID,
    "revision": BGE_REVISION,
    "local_path": str(BGE_LOCAL_PATH),
    "downloads": False,
    "model_inference_api_calls": 0,
    "local_model_pairs": 0,
}
BGE_LOCAL_READY = False

try:
    receipts = []

    for name, (expected_size, hash_type, expected_hash) in BGE_EXPECTED_FILES.items():
        path = BGE_LOCAL_PATH / name
        print(f"بررسی: {name}", flush=True)

        if not path.is_file():
            raise RuntimeError(f"فایل موجود نیست یا لینک آن ناقص است: {path}")

        actual_size = path.stat().st_size
        if actual_size != expected_size:
            raise RuntimeError(
                f"اندازهٔ نامعتبر برای {name}: "
                f"{actual_size}؛ مقدار مورد انتظار: {expected_size}"
            )

        sha256 = hashlib.sha256()
        verifier = (
            hashlib.sha256()
            if hash_type == "sha256"
            else hashlib.sha1()
        )

        if hash_type == "git_sha1":
            verifier.update(f"blob {actual_size}\0".encode("ascii"))

        with path.open("rb") as stream:
            for block in iter(lambda: stream.read(8 * 1024 * 1024), b""):
                verifier.update(block)
                sha256.update(block)

        if verifier.hexdigest() != expected_hash:
            raise RuntimeError(f"هش نامعتبر: {name}؛ فایل قابل استفاده نیست.")

        receipts.append({
            "name": name,
            "bytes": actual_size,
            "sha256": sha256.hexdigest(),
            "validation": "PASS",
        })
        print(f"PASS: {name}", flush=True)

    config = json.loads(
        (BGE_LOCAL_PATH / "config.json").read_text(encoding="utf-8")
    )

    if (
        config.get("model_type") != "xlm-roberta"
        or config.get("architectures")
        != ["XLMRobertaForSequenceClassification"]
        or config.get("auto_map")
    ):
        raise RuntimeError("پیکربندی مدل با قرارداد مورد انتظار مطابقت ندارد.")

    C015.update({
        "status": "PINNED_FILES_VERIFIED_NOT_EXECUTED",
        "files": receipts,
        "runtime_memory_and_speed_verified": False,
        "quality_improvement_verified": False,
        "next_step": "BOUNDED_CPU_TIMING_AND_MEMORY_PILOT",
    })

    # استفاده از سازوکار ثبت گزارش نوت‌بوک، اگر در کرنل موجود باشد
    notebook_save = globals().get("save")
    if callable(notebook_save):
        C015["report"] = notebook_save(
            "C015.pinned_model_asset", C015.copy()
        )

    BGE_LOCAL_READY = True
    print(json.dumps(C015, ensure_ascii=False, indent=2))

except Exception as exc:
    BGE_LOCAL_READY = False
    C015.update({
        "status": "LOCAL_VERIFICATION_FAILED",
        "error": str(exc),
    })
    print(json.dumps(C015, ensure_ascii=False, indent=2))
    raise
```

<h3>C016 — نسخهٔ ۲: آزمون کوتاه اجرای محلی مدل BGE</h3>

<p><strong>هدف:</strong> اندازه‌گیری سرعت و حافظه، پیش از آزمایش اصلی کیفیت.</p>

<ul>
<li>ابتدا متن سؤال‌ها و قطعه‌ها در وضعیت موجود نوت‌بوک بررسی می‌شود.</li>
<li>حداکثر سه جفت واقعی، با اولویت منابع متفاوت، پردازش می‌شود.</li>
<li>اجرا روی CPU، با batch برابر ۱ و حداکثر طول ۵۱۲ توکن است.</li>
<li>دانلود، API و تغییر در baseline انجام نمی‌شود.</li>
<li>این آزمون فقط امکان اجرای مدل را می‌سنجد؛ پذیرش کیفیت نیست.</li>
</ul>

</div>

```python
# C016 v3 — Verified project passages + bounded local BGE CPU pilot

import os
import sys
import time
import json
import math
import hashlib
import resource
from pathlib import Path

os.environ["TOKENIZERS_PARALLELISM"] = "false"

C016 = {
    "version": "NB22E_BGE_CPU_PILOT_V3",
    "status": "PREPARING",
    "downloads": False,
    "model_inference_api_calls": 0,
    "local_model_pairs": 0,
}


def c016_resolve_passages(chunk_ids):
    relative_path = (
        "data/phase2/chunks/nb11_benchmark_v1/"
        "baseline_phase1_char_safe_v1.jsonl"
    )
    expected_sha256 = (
        "3891c4ecad9332ba753cd605bbccd99da9d"
        "063f64934a30ee39e495349cffb50"
    )

    project_root = globals().get("ROOT")
    if project_root is None:
        raise RuntimeError("ROOT پروژه در کرنل موجود نیست.")

    root = Path(project_root).expanduser().resolve()
    path = (root / relative_path).resolve()

    if not path.is_relative_to(root):
        raise RuntimeError("مسیر قطعه‌ها خارج از پروژه است.")
    if not path.is_file():
        raise FileNotFoundError(f"فایل قطعه‌ها پیدا نشد: {path}")

    print("بررسی هش فایل قطعه‌های پروژه…", flush=True)

    # Hash verification precedes parsing or using passage content.
    hasher = hashlib.sha256()
    with path.open("rb") as stream:
        for block in iter(lambda: stream.read(8 * 1024 * 1024), b""):
            hasher.update(block)

    if hasher.hexdigest() != expected_sha256:
        raise RuntimeError(
            "هش فایل قطعه‌ها با نسخهٔ مبنا تطابق ندارد؛ "
            "مدل بارگذاری نشد."
        )

    needed = set(chunk_ids)
    found = {}

    with path.open("r", encoding="utf-8") as stream:
        for line_number, line in enumerate(stream, start=1):
            if not line.strip():
                continue

            record = json.loads(line)
            if not isinstance(record, dict):
                raise RuntimeError(f"رکورد نامعتبر در خط {line_number}")

            cid = record.get("chunk_id")
            if cid not in needed:
                continue
            if cid in found:
                raise RuntimeError(f"شناسهٔ قطعه تکراری: {cid}")

            text = record.get("content")
            if not isinstance(text, str) or not text.strip():
                raise RuntimeError(f"متن قطعه نامعتبر است: {cid}")
            found[cid] = text

    missing = needed - found.keys()
    if missing:
        raise RuntimeError(f"قطعه‌های پیدا‌نشده: {sorted(missing)}")

    print(f"PASS: فایل و متن {len(found)} قطعه تأیید شد.", flush=True)
    return found, relative_path, expected_sha256


try:
    # ---------- Preconditions ----------
    asset = globals().get("C015")
    if (
        not globals().get("BGE_LOCAL_READY", False)
        or not isinstance(asset, dict)
        or asset.get("status") != "PINNED_FILES_VERIFIED_NOT_EXECUTED"
    ):
        raise RuntimeError("ابتدا C015 اصلاحی باید موفق اجرا شود.")

    model_id = "BAAI/bge-reranker-v2-m3"
    revision = "953dc6f6f85a1b2dbfca4c34a2796e7dde08d41e"

    if (
        asset.get("model_id") != model_id
        or asset.get("revision") != revision
    ):
        raise RuntimeError("مشخصات C015 با مدل مورد انتظار تطابق ندارد.")

    model_path = Path(asset["local_path"]).expanduser().resolve()
    if not model_path.is_dir():
        raise FileNotFoundError(f"پوشهٔ مدل پیدا نشد: {model_path}")

    lab = globals().get("LAB")
    if not isinstance(lab, dict):
        raise RuntimeError("LAB در کرنل موجود نیست.")

    required_keys = ("pilot", "questions", "pools", "source_types")
    missing_keys = [key for key in required_keys if key not in lab]
    if missing_keys:
        raise RuntimeError(f"کلیدهای موردنیاز LAB موجود نیست: {missing_keys}")

    # ---------- Select up to three real pairs ----------
    # Selection uses source type and pool order, not gold labels.
    samples = []
    seen_sources = set()

    for qid in lab["pilot"]:
        source_types = lab["source_types"][qid]
        if isinstance(source_types, str):
            source_types = [source_types]

        source = "|".join(sorted(source_types))
        if source in seen_sources:
            continue

        pool = lab["pools"][qid]
        if not pool:
            continue

        question = lab["questions"][qid]
        if not isinstance(question, str) or not question.strip():
            raise RuntimeError(f"متن سؤال نامعتبر: {qid}")

        samples.append({
            "query_id": qid,
            "chunk_id": pool[0],
            "source": source,
            "question": question,
        })
        seen_sources.add(source)

        if len(samples) == 3:
            break

    if not samples:
        raise RuntimeError("هیچ جفت واقعی برای آزمون پیدا نشد.")

    passages, passage_path, passage_sha256 = c016_resolve_passages(
        [sample["chunk_id"] for sample in samples]
    )

    # ---------- Load locally or reuse matching runtime ----------
    import torch
    from transformers import (
        AutoTokenizer,
        AutoModelForSequenceClassification,
    )

    threads = min(4, os.cpu_count() or 1)
    torch.set_num_threads(threads)

    runtime = globals().get("BGE_RUNTIME")
    reusable = (
        isinstance(runtime, dict)
        and runtime.get("model_id") == model_id
        and runtime.get("revision") == revision
        and runtime.get("device") == "cpu"
        and runtime.get("dtype") == "float32"
        and runtime.get("local_path") == str(model_path)
        and runtime.get("model") is not None
        and runtime.get("tokenizer") is not None
    )

    if reusable:
        model = runtime["model"]
        tokenizer = runtime["tokenizer"]
        load_seconds = 0.0
        print("استفاده از مدل موجود در حافظه.", flush=True)
    else:
        print(json.dumps({
            "stage": "LOCAL_MODEL_LOADING",
            "model_id": model_id,
            "device": "cpu",
            "dtype": "float32",
            "threads": threads,
            "planned_pairs": len(samples),
            "downloads": False,
            "model_inference_api_calls": 0,
        }, ensure_ascii=False), flush=True)

        started = time.perf_counter()

        tokenizer = AutoTokenizer.from_pretrained(
            str(model_path),
            local_files_only=True,
            trust_remote_code=False,
        )
        model = AutoModelForSequenceClassification.from_pretrained(
            str(model_path),
            local_files_only=True,
            trust_remote_code=False,
            use_safetensors=True,
            torch_dtype=torch.float32,
        )
        model.to("cpu")
        model.eval()

        load_seconds = time.perf_counter() - started

    model.eval()

    BGE_RUNTIME = {
        "model": model,
        "tokenizer": tokenizer,
        "model_id": model_id,
        "revision": revision,
        "local_path": str(model_path),
        "device": "cpu",
        "dtype": "float32",
        "threads": threads,
        "batch_size": 1,
        "max_length": 512,
    }

    print(
        f"زمان بارگذاری در این اجرا: {load_seconds:.1f} ثانیه",
        flush=True,
    )

    # ---------- Small timing pilot ----------
    results = []
    inference_started = time.perf_counter()
    between_pair_limit_seconds = 180

    for sample in samples:
        # Soft limit: checked between pairs, not during a forward pass.
        if (
            results
            and time.perf_counter() - inference_started
            >= between_pair_limit_seconds
        ):
            break

        print(
            f"پردازش جفت {len(results) + 1}/{len(samples)}"
            f" | منبع: {sample['source']}",
            flush=True,
        )

        started = time.perf_counter()
        encoded = tokenizer(
            sample["question"],
            passages[sample["chunk_id"]],
            return_tensors="pt",
            truncation=True,
            max_length=512,
            padding=False,
        )
        input_tokens = int(encoded["input_ids"].shape[1])

        with torch.inference_mode():
            logits = model(**encoded).logits

        if logits.numel() != 1:
            raise RuntimeError(
                f"شکل خروجی غیرمنتظره: {tuple(logits.shape)}"
            )

        score = float(logits.item())
        if not math.isfinite(score):
            raise RuntimeError("امتیاز مدل متناهی نیست.")

        row = {
            "query_id": sample["query_id"],
            "chunk_id": sample["chunk_id"],
            "source": sample["source"],
            "input_tokens": input_tokens,
            "seconds": time.perf_counter() - started,
            "raw_score": score,
        }
        results.append(row)
        C016["local_model_pairs"] = len(results)
        print(json.dumps(row, ensure_ascii=False), flush=True)

    # ---------- Report ----------
    peak = resource.getrusage(resource.RUSAGE_SELF).ru_maxrss
    peak_bytes = peak if sys.platform == "darwin" else peak * 1024
    mean_seconds = sum(row["seconds"] for row in results) / len(results)

    C016.update({
        "status": (
            "PILOT_COMPLETE"
            if len(results) == len(samples)
            else "PILOT_STOPPED_AT_BETWEEN_PAIR_TIME_LIMIT"
        ),
        "model_id": model_id,
        "revision": revision,
        "passage_binding": {
            "path": passage_path,
            "file_bytes_sha256": passage_sha256,
        },
        "device": "cpu",
        "dtype": "float32",
        "threads": threads,
        "batch_size": 1,
        "max_length": 512,
        "runtime_reused": reusable,
        "load_seconds": load_seconds,
        "mean_seconds_per_pair": mean_seconds,
        "process_lifetime_peak_rss_gib": peak_bytes / 1024**3,
        "memory_scope": "WHOLE_KERNEL_LIFETIME_PEAK_NOT_MODEL_ONLY",
        "rough_2400_pairs_hours": mean_seconds * 2400 / 3600,
        "estimate_limit": "SMALL_SAMPLE_ONLY_NOT_VALIDATED_FORECAST",
        "between_pair_limit_seconds": between_pair_limit_seconds,
        "limit_scope": "EXCLUDES_LOADING_AND_CANNOT_INTERRUPT_ACTIVE_PAIR",
        "samples": results,
        "quality_improvement_verified": False,
        "automatic_replacement": False,
    })

    notebook_save = globals().get("save")
    if callable(notebook_save):
        C016["report"] = notebook_save(
            "C016.bge_cpu_pilot", C016.copy()
        )

    print(json.dumps(C016, ensure_ascii=False, indent=2))

except KeyboardInterrupt:
    C016.update({"status": "INTERRUPTED_BY_USER"})
    print(json.dumps(C016, ensure_ascii=False, indent=2))
    raise

except Exception as exc:
    C016.update({
        "status": "FAILED",
        "error": str(exc),
    })
    print(json.dumps(C016, ensure_ascii=False, indent=2))
    raise
```

<h3>C017 — نسخهٔ ۱: مقایسهٔ BGE با baseline روی ۶۰ سؤال</h3>

<p>هدف: بررسی اثر مدل جدید بر رتبه‌بندی ۲۰ نتیجهٔ اول baseline،
بدون تغییر مجموعهٔ بازیابی یا ترکیب امتیاز مدل‌ها.</p>

<li>همان ۶۰ سؤال آزمایش قبلی؛ این مجموعه برای پذیرش مستقل نیست.</li>

<li>حداکثر ۱۲۰۰ جفت؛ ذخیرهٔ تدریجی و امکان ادامه پس از توقف.</li>

<li>برای Hit@10 حد مطلق تعیین نشده؛ افت نسبت به baseline بررسی می‌شود.</li>
<li>baseline به‌صورت خودکار جایگزین نمی‌شود.</li>
</ul>

```python
# C017 v1 — BGE reranks baseline Top20; resumable local scoring.
# No downloads, no inference API, no automatic baseline replacement.

import json
import math
import time
import hashlib
import os
import tempfile
from pathlib import Path

C017 = {
    "version": "NB22E_BGE_TOP20_V1",
    "status": "PREPARING",
    "downloads": False,
    "new_api_calls": 0,
    "automatic_replacement": False,
    "release_acceptance_granted": False,
}

def c017_canonical(value):
    return json.dumps(
        value, ensure_ascii=False, sort_keys=True,
        separators=(",", ":"), allow_nan=False,
    ).encode("utf-8")

def c017_hash(value):
    return hashlib.sha256(c017_canonical(value)).hexdigest()

def c017_read_pinned(relative, expected):
    path = (Path(ROOT).resolve() / relative).resolve()
    if not path.is_relative_to(Path(ROOT).resolve()):
        raise RuntimeError("مسیر خارج از پروژه است.")
    raw = path.read_bytes()
    if hashlib.sha256(raw).hexdigest() != expected:
        raise RuntimeError(f"هش فایل تطابق ندارد: {relative}")
    return [
        json.loads(line)
        for line in raw.decode("utf-8").splitlines()
        if line.strip()
    ]

def c017_atomic_json(path, value):
    payload = c017_canonical(value)
    fd, temporary = tempfile.mkstemp(
        prefix=".pending_", dir=str(path.parent)
    )
    try:
        with os.fdopen(fd, "wb") as stream:
            stream.write(payload)
            stream.flush()
            os.fsync(stream.fileno())
        os.replace(temporary, path)
    finally:
        if os.path.exists(temporary):
            os.unlink(temporary)

def c017_metrics(order, relevant):
    top = order[:10]
    hits = [int(cid in relevant) for cid in top]
    dcg = sum(hit / math.log2(i + 2) for i, hit in enumerate(hits))
    idcg = sum(
        1 / math.log2(i + 2)
        for i in range(min(10, len(relevant)))
    )
    return {
        "recall_at_10": sum(hits) / len(relevant),
        "ndcg_at_10": dcg / idcg,
        "hit_at_10": float(any(hits)),
    }

def c017_mean(rows):
    return {
        key: sum(row[key] for row in rows) / len(rows)
        for key in ("recall_at_10", "ndcg_at_10", "hit_at_10")
    }

try:
    import torch
    from IPython.display import display, HTML

    if C016.get("status") != "PILOT_COMPLETE":
        raise RuntimeError("ابتدا C016 باید کامل شود.")
    if not BGE_LOCAL_READY:
        raise RuntimeError("تأیید فایل‌های مدل موجود نیست.")

    runtime = globals().get("BGE_RUNTIME")
    if not isinstance(runtime, dict):
        raise RuntimeError("مدل در حافظه موجود نیست؛ C016 را اجرا کن.")
    if (
        runtime.get("model_id") != "BAAI/bge-reranker-v2-m3"
        or runtime.get("revision")
        != "953dc6f6f85a1b2dbfca4c34a2796e7dde08d41e"
        or runtime.get("device") != "cpu"
        or runtime.get("dtype") != "float32"
    ):
        raise RuntimeError("قرارداد اجرای مدل تطابق ندارد.")

    qids = list(LAB["pilot"])
    if len(qids) != 60 or len(set(qids)) != 60:
        raise RuntimeError("این سل دقیقاً به همان مجموعهٔ ۶۰ سؤال نیاز دارد.")

    model = runtime["model"]
    tokenizer = runtime["tokenizer"]
    model.eval()
    torch.set_num_threads(4)

    print("بررسی فایل‌های قطعه‌ها و Golden…", flush=True)

    chunk_rows = c017_read_pinned(
        "data/phase2/chunks/nb11_benchmark_v1/"
        "baseline_phase1_char_safe_v1.jsonl",
        "3891c4ecad9332ba753cd605bbccd99da9d"
        "063f64934a30ee39e495349cffb50",
    )
    golden_rows = c017_read_pinned(
        "artifacts/phase2/golden_dataset_v2/golden_dataset_v2.jsonl",
        "b9ca18e710686ddca48a25c637840b716"
        "adb306717cafd8ec7c96ce9e182e6f2",
    )

    chunks = {}
    locator_chunks = {}
    for row in chunk_rows:
        cid = row["chunk_id"]
        if cid in chunks:
            raise RuntimeError(f"قطعهٔ تکراری: {cid}")
        chunks[cid] = row
        for member in row.get("member_locators", []):
            key = c017_canonical(member["locator"])
            locator_chunks.setdefault(key, set()).add(cid)

    golden = {}
    for row in golden_rows:
        qid = row["candidate_id"]
        if qid in golden:
            raise RuntimeError(f"سؤال تکراری: {qid}")
        golden[qid] = row

    relevant = {}
    baseline_orders = {}
    before_by_query = {}

    for qid in qids:
        g = golden[qid]
        if g["evaluation_content"]["question"] != LAB["questions"][qid]:
            raise RuntimeError(f"متن سؤال با Golden تطابق ندارد: {qid}")

        allowed_sources = set(g["source_types"])
        gold_ids = set()

        for evidence in g["evidence_locators"]:
            matches = locator_chunks.get(
                c017_canonical(evidence["source_locator"]), set()
            )
            gold_ids.update(
                cid for cid in matches
                if chunks[cid]["source_type"] in allowed_sources
            )

        if not gold_ids:
            raise RuntimeError(f"شاهد قابل اتصال برای سؤال پیدا نشد: {qid}")

        order = list(LAB["incumbent"][qid])
        if (
            len(order) < 20
            or len(set(order)) != len(order)
            or any(cid not in chunks for cid in order)
        ):
            raise RuntimeError(f"ترتیب baseline نامعتبر است: {qid}")

        relevant[qid] = gold_ids
        baseline_orders[qid] = order
        before_by_query[qid] = c017_metrics(order, gold_ids)

    before = c017_mean(list(before_by_query.values()))

    # Confirm that evaluation reproduces the previously reported baseline.
    previous = LAB["results"].get("baseline_D_60", {}).get("metrics")
    if not isinstance(previous, dict):
        raise RuntimeError("معیارهای baseline_D_60 در کرنل موجود نیست.")
    for key in before:
        if abs(before[key] - float(previous[key])) > 1e-9:
            raise RuntimeError(
                f"بازسازی معیار baseline تطابق ندارد: {key}. "
                "هیچ پردازش مدل شروع نشد."
            )

    contract = {
        "version": "BGE_TOP20_CPU_FLOAT32_BATCH1_V1",
        "model_id": runtime["model_id"],
        "revision": runtime["revision"],
        "asset_files": C015["files"],
        "torch": torch.__version__,
        "threads": 4,
        "max_length": 512,
        "padding": False,
        "truncation": True,
        "batch_size": 1,
        "score": "raw_single_logit",
    }
    contract_hash = c017_hash(contract)

    cache_dir = (
        Path(ROOT).resolve()
        / "artifacts/phase2/quality_improvement"
        / "NB22E.bge_top20_cache_v1"
        / contract_hash
    )
    cache_dir.mkdir(parents=True, exist_ok=True)

    started = time.perf_counter()
    new_pairs = 0
    cached_pairs = 0
    processed_pairs = 0
    after_by_query = {}
    final_orders = {}

    print(json.dumps({
        "stage": "BGE_TOP20_SCORING",
        "queries": 60,
        "maximum_pairs": 1200,
        "contract_sha256": contract_hash,
        "downloads": False,
        "new_api_calls": 0,
    }), flush=True)

    for query_index, qid in enumerate(qids, start=1):
        question = LAB["questions"][qid]
        top20 = baseline_orders[qid][:20]
        scores = {}

        for cid in top20:
            passage = chunks[cid]["content"]
            if not isinstance(passage, str) or not passage.strip():
                raise RuntimeError(f"متن قطعه نامعتبر: {cid}")

            identity = {
                "contract_sha256": contract_hash,
                "query_id": qid,
                "chunk_id": cid,
                "question_sha256": hashlib.sha256(
                    question.encode("utf-8")
                ).hexdigest(),
                "passage_sha256": hashlib.sha256(
                    passage.encode("utf-8")
                ).hexdigest(),
            }
            cache_path = cache_dir / (c017_hash(identity) + ".json")

            if cache_path.exists():
                record = json.loads(cache_path.read_text(encoding="utf-8"))
                body = record["body"]
                if (
                    record["body_sha256"] != c017_hash(body)
                    or body["identity"] != identity
                ):
                    raise RuntimeError(f"کش نامعتبر: {cache_path}")
                score = float(body["score"])
                cached_pairs += 1
            else:
                encoded = tokenizer(
                    question, passage,
                    return_tensors="pt",
                    truncation=True,
                    max_length=512,
                    padding=False,
                )
                with torch.inference_mode():
                    logits = model(**encoded).logits

                if logits.numel() != 1:
                    raise RuntimeError("شکل خروجی مدل نامعتبر است.")
                score = float(logits.item())
                if not math.isfinite(score):
                    raise RuntimeError("امتیاز مدل متناهی نیست.")

                body = {
                    "identity": identity,
                    "score": score,
                    "input_tokens": int(encoded["input_ids"].shape[1]),
                }
                c017_atomic_json(cache_path, {
                    "body": body,
                    "body_sha256": c017_hash(body),
                })
                new_pairs += 1

            if not math.isfinite(score):
                raise RuntimeError("امتیاز کش متناهی نیست.")
            scores[cid] = score
            processed_pairs += 1

        # Stable ties preserve baseline order. Tail stays in its original order.
        positions = {cid: i for i, cid in enumerate(top20)}
        reranked = sorted(
            top20, key=lambda cid: (-scores[cid], positions[cid])
        )
        order = reranked + baseline_orders[qid][20:]
        final_orders[qid] = order
        after_by_query[qid] = c017_metrics(order, relevant[qid])

        print(json.dumps({
            "queries_completed": query_index,
            "queries_total": 60,
            "pairs_processed": processed_pairs,
            "new_pairs": new_pairs,
            "cached_pairs": cached_pairs,
            "elapsed_minutes": round(
                (time.perf_counter() - started) / 60, 2
            ),
        }), flush=True)

    after = c017_mean(list(after_by_query.values()))
    per_source = {}

    for source in ("book", "transcript", "website"):
        ids = [
            qid for qid in qids
            if source in golden[qid]["source_types"]
        ]
        if ids:
            per_source[source] = {
                "queries": len(ids),
                "before": c017_mean([before_by_query[q] for q in ids]),
                "after": c017_mean([after_by_query[q] for q in ids]),
            }

    no_regression = all(
        after[key] >= before[key] - 1e-12 for key in before
    )

    C017.update({
        "status": "COMPLETE",
        "query_count": 60,
        "method": "BGE_RERANK_BASELINE_TOP20",
        "contract": contract,
        "contract_sha256": contract_hash,
        "before": before,
        "after": after,
        "delta_percentage_points": {
            key: (after[key] - before[key]) * 100 for key in before
        },
        "per_source": per_source,
        "new_model_pairs": new_pairs,
        "cached_pairs": cached_pairs,
        "wall_seconds": time.perf_counter() - started,
        "aggregate_no_regression": no_regression,
        "exploratory_candidate": (
            no_regression
            and after["ndcg_at_10"] > before["ndcg_at_10"] + 1e-12
        ),
        "absolute_quality_targets_met": (
            after["recall_at_10"] >= 0.85
            and after["ndcg_at_10"] >= 0.75
        ),
        "scope": "SEEN_60_DEV_QUERIES_NOT_INDEPENDENT_ACCEPTANCE",
    })

    C017["orders_artifact"] = save(
        "C017.bge_top20_orders",
        {
            "contract_sha256": contract_hash,
            "orders": final_orders,
            "before_by_query": before_by_query,
            "after_by_query": after_by_query,
        },
    )
    C017["report"] = save("C017.bge_top20_report", C017.copy())

    rows = []
    for key, label, target in (
        ("recall_at_10", "Recall@10", 0.85),
        ("ndcg_at_10", "nDCG@10", 0.75),
        ("hit_at_10", "Hit@10", None),
    ):
        threshold = "تعیین نشده" if target is None else f"{target:.0%}"
        accepted = "✅" if target is not None and after[key] >= target else ""
        rows.append(
            f"<tr><td>{label}</td>"
            f"<td>{before[key]:.2%}</td>"
            f"<td>{after[key]:.2%}</td>"
            f"<td>{100 * (after[key] - before[key]):+.2f}</td>"
            f"<td>{threshold}</td><td>{accepted}</td></tr>"
        )

    display(HTML(
        '<div dir="rtl" style="text-align:right">'
        "<p>مقایسهٔ آزمایشی روی همان ۶۰ سؤال دیده‌شده؛ "
        "پذیرش مستقل نیست. تیک فقط رسیدن به حد عددی همان معیار است.</p>"
        '<table style="text-align:center">'
        "<tr><th>معیار</th><th>قبل</th><th>بعد</th>"
        "<th>تغییر؛ واحد درصد</th><th>حد پذیرش</th><th>قبول</th></tr>"
        + "".join(rows) + "</table></div>"
    ))
    print(json.dumps(C017, ensure_ascii=False, indent=2))

except KeyboardInterrupt:
    C017.update({
        "status": "INTERRUPTED_RESUMABLE",
        "message": "برای ادامه همین سل را دوباره اجرا کن؛ کش حفظ شده است.",
    })
    print(json.dumps(C017, ensure_ascii=False, indent=2))
    raise

except Exception as exc:
    C017.update({"status": "FAILED", "error": str(exc)})
    print(json.dumps(C017, ensure_ascii=False, indent=2))
    raise
```

<h3>C018 — نسخهٔ ۱: بررسی ظرفیت گسترش ورودی BGE</h3>

<p>هدف: مشخص‌کردن اینکه افزایش تعداد نتایج ورودی به مدل،
از نظر حضور شاهدهای صحیح چقدر ظرفیت بهبود دارد.</p>

<ul>
<li>مقایسهٔ Top20، Top40 و Top60 از ترتیب اصلی baseline.</li>
<li>محاسبهٔ سقف نظری Recall@10؛ نه کیفیت واقعی مدل.</li>
<li>انتخاب کوچک‌ترین عمقی که سقف آن به هدف ۸۵٪ برسد.</li>
<li>رسیدن سقف به هدف، تضمین رسیدن مدل به هدف نیست.</li>
<li>بدون اجرای مدل، دانلود، API یا تغییر baseline.</li>
<li>همان ۶۰ سؤال دیده‌شده؛ پذیرش مستقل نیست.</li>
</ul>

</div>

```python
# C018 v1 — Offline candidate-depth ceilings; no model inference.

import json
import math
from IPython.display import display, HTML

C018 = {
    "version": "NB22E_BGE_DEPTH_CEILING_V1",
    "status": "PREPARING",
    "new_model_pairs": 0,
    "new_api_calls": 0,
    "downloads": False,
    "automatic_replacement": False,
    "release_acceptance_granted": False,
}

try:
    if C017.get("status") != "COMPLETE":
        raise RuntimeError("ابتدا C017 باید کامل شده باشد.")

    required = (
        "qids", "relevant", "baseline_orders",
        "golden", "before_by_query", "after_by_query",
    )
    missing = [name for name in required if name not in globals()]
    if missing:
        raise RuntimeError(
            f"متغیرهای اجرای C017 موجود نیستند: {missing}. "
            "فعلاً هیچ محاسبهٔ مدل انجام نشد."
        )

    # Use the exact cohort from the completed C017 evaluation.
    audit_ids = list(qids)
    if len(audit_ids) != 60 or len(set(audit_ids)) != 60:
        raise RuntimeError("مجموعهٔ ارزیابی باید دقیقاً ۶۰ سؤال باشد.")
    if set(audit_ids) != set(LAB["pilot"]):
        raise RuntimeError("مجموعهٔ جاری LAB با C017 تطابق ندارد.")

    for qid in audit_ids:
        if not relevant[qid]:
            raise RuntimeError(f"مجموعهٔ شاهد خالی است: {qid}")

        order = baseline_orders[qid]
        if len(order) < 60 or len(set(order)) != len(order):
            raise RuntimeError(f"ترتیب baseline کوتاه یا تکراری است: {qid}")
        if order != list(LAB["incumbent"][qid]):
            raise RuntimeError(f"ترتیب baseline پس از C017 تغییر کرده: {qid}")

    # Ensure the in-memory evaluation still reproduces the saved result.
    for name, values in (
        ("before", before_by_query),
        ("after", after_by_query),
    ):
        for metric in ("recall_at_10", "ndcg_at_10", "hit_at_10"):
            actual = sum(values[q][metric] for q in audit_ids) / 60
            if abs(actual - C017[name][metric]) > 1e-9:
                raise RuntimeError(
                    f"معیارهای حافظه با C017 تطابق ندارند: {name}/{metric}"
                )

    target = 0.85
    depths = (20, 40, 60)
    rows = []
    per_query = []

    for depth in depths:
        ceilings = []
        hits = []
        oracle_ndcgs = []

        for qid in audit_ids:
            gold = relevant[qid]
            available = len(set(baseline_orders[qid][:depth]) & gold)

            # At most 10 relevant chunks can appear in the final Top10.
            best_top10_count = min(10, available)
            recall_ceiling = best_top10_count / len(gold)
            ideal_dcg = sum(
                1 / math.log2(i + 2)
                for i in range(min(10, len(gold)))
            )
            best_dcg = sum(
                1 / math.log2(i + 2)
                for i in range(best_top10_count)
            )

            ceilings.append(recall_ceiling)
            hits.append(float(available > 0))
            oracle_ndcgs.append(best_dcg / ideal_dcg)

            per_query.append({
                "query_id": qid,
                "depth": depth,
                "available_gold_chunks": available,
                "total_gold_chunks": len(gold),
                "recall_at_10_ceiling": recall_ceiling,
            })

        per_source = {}
        for source in ("book", "transcript", "website"):
            indices = [
                i for i, q in enumerate(audit_ids)
                if source in golden[q]["source_types"]
            ]
            if indices:
                per_source[source] = {
                    "queries": len(indices),
                    "recall_at_10_ceiling": (
                        sum(ceilings[i] for i in indices) / len(indices)
                    ),
                }

        rows.append({
            "depth": depth,
            "recall_at_10_ceiling": sum(ceilings) / 60,
            "ndcg_at_10_ceiling": sum(oracle_ndcgs) / 60,
            "hit_at_10_ceiling": sum(hits) / 60,
            "maximum_additional_pairs_over_top20": (depth - 20) * 60,
            "per_source": per_source,
        })

    baseline_recall = C017["before"]["recall_at_10"]
    bge_recall = C017["after"]["recall_at_10"]
    qualifying = [
        row["depth"] for row in rows
        if row["recall_at_10_ceiling"] >= target - 1e-12
    ]
    selected = min(qualifying) if qualifying else None

    if selected == 20:
        next_step = (
            "TOP20_ALREADY_HAS_THEORETICAL_TARGET_CAPACITY; "
            "DO_NOT_EXPAND_AUTOMATICALLY"
        )
    elif selected is not None:
        next_step = (
            f"TEST_BGE_TOP{selected}_REUSING_TOP20_CACHE; "
            "CEILING_IS_NOT_A_QUALITY_GUARANTEE"
        )
    else:
        next_step = (
            "NO_TESTED_DEPTH_HAS_TARGET_CAPACITY; "
            "DO_NOT_START_LARGER_BGE_RUN_AUTOMATICALLY"
        )

    C018.update({
        "status": "COMPLETE",
        "query_count": 60,
        "baseline_recall_at_10": baseline_recall,
        "bge_top20_recall_at_10": bge_recall,
        "recall_target": target,
        "depths": rows,
        "smallest_depth_with_target_capacity": selected,
        "selection_scope": "ORACLE_CAPACITY_ONLY_NOT_MODEL_PERFORMANCE",
        "next_step": next_step,
        "scope": "SEEN_60_DEV_QUERIES_NOT_INDEPENDENT_ACCEPTANCE",
    })

    C018["cases_artifact"] = save(
        "C018.depth_ceiling_cases",
        {"queries": per_query},
    )
    C018["report"] = save("C018.depth_ceiling_report", C018.copy())

    table_rows = []
    for row in rows:
        ceiling = row["recall_at_10_ceiling"]
        tick = "✅" if ceiling >= target - 1e-12 else ""
        table_rows.append(
            f"<tr><td>Top{row['depth']}</td>"
            f"<td>{baseline_recall:.2%}</td>"
            f"<td>{bge_recall:.2%}</td>"
            f"<td>{ceiling:.2%}</td>"
            f"<td>{target:.0%}</td><td>{tick}</td>"
            f"<td>{row['maximum_additional_pairs_over_top20']}</td></tr>"
        )

    display(HTML(
        '<div dir="rtl" style="text-align:right">'
        "<p><strong>سقف نظری است؛ نتیجهٔ واقعی اجرای BGE نیست.</strong> "
        "تیک فقط ظرفیت نظری رسیدن به هدف را نشان می‌دهد.</p>"
        '<table style="text-align:center">'
        "<tr><th>ورودی</th><th>قبل: baseline</th>"
        "<th>بعد: BGE Top20 فعلی</th><th>سقف Recall@10</th>"
        "<th>هدف</th><th>ظرفیت کافی</th>"
        "<th>حداکثر جفت اضافه</th></tr>"
        + "".join(table_rows)
        + "</table></div>"
    ))
    print(json.dumps(C018, ensure_ascii=False, indent=2))

except Exception as exc:
    C018.update({"status": "FAILED", "error": str(exc)})
    print(json.dumps(C018, ensure_ascii=False, indent=2))
    raise
```

<h3>C019 — نسخهٔ ۱: ارزیابی BGE روی Top40</h3>
<p>هدف: افزایش پوشش شاهدها با گسترش ورودی مدل از ۲۰ به ۴۰ نتیجهٔ baseline.</p>
<ul>
<li>همان ۶۰ سؤال؛ استفاده از امتیازهای ذخیره‌شدهٔ Top20.</li>
<li>حداکثر ۱۲۰۰ جفت جدید؛ ذخیرهٔ هر جفت و امکان ادامه پس از توقف.</li>
<li>مقایسهٔ baseline، BGE Top20 و BGE Top40.</li>
<li>بدون دانلود، API و جایگزینی خودکار baseline.</li>
<li>سقف نظری ۸۸٫۶۱٪ تضمین عملکرد واقعی نیست؛ پذیرش مستقل انجام نمی‌شود.</li>
</ul>
</div>

```python
# C019 v1 — Extend BGE Top20 to Top40, preserving the score contract/cache.

import json
import math
import time
import hashlib
import torch
from pathlib import Path
from IPython.display import display, HTML

C019 = {
    "version": "NB22E_BGE_TOP40_V1",
    "status": "PREPARING",
    "downloads": False,
    "new_api_calls": 0,
    "automatic_replacement": False,
    "release_acceptance_granted": False,
}

try:
    if C017.get("status") != "COMPLETE":
        raise RuntimeError("C017 کامل نیست.")
    if (
        C018.get("status") != "COMPLETE"
        or C018.get("smallest_depth_with_target_capacity") != 40
    ):
        raise RuntimeError("انتخاب Top40 توسط C018 تأیید نشده است.")

    # Reuse C017's exact scoring contract and helper functions.
    required = (
        "contract", "contract_hash", "cache_dir", "qids", "chunks",
        "relevant", "baseline_orders", "golden",
        "c017_hash", "c017_atomic_json", "c017_metrics", "c017_mean",
    )
    missing = [name for name in required if name not in globals()]
    if missing:
        raise RuntimeError(f"متغیرهای C017 موجود نیستند: {missing}")

    if c017_hash(contract) != C017["contract_sha256"]:
        raise RuntimeError("قرارداد امتیازدهی تغییر کرده است.")

    runtime = BGE_RUNTIME
    if (
        runtime["model_id"] != contract["model_id"]
        or runtime["revision"] != contract["revision"]
        or runtime["device"] != "cpu"
        or runtime["dtype"] != "float32"
        or torch.__version__ != contract["torch"]
    ):
        raise RuntimeError("مدل جاری با قرارداد کش تطابق ندارد.")

    ids40 = list(qids)
    if (
        len(ids40) != 60
        or len(set(ids40)) != 60
        or set(ids40) != set(LAB["pilot"])
    ):
        raise RuntimeError("مجموعهٔ ۶۰ سؤال تغییر کرده است.")

    model40 = runtime["model"]
    tokenizer40 = runtime["tokenizer"]
    model40.eval()
    torch.set_num_threads(contract["threads"])

    def c019_pair_identity(qid, cid):
        return {
            "contract_sha256": C017["contract_sha256"],
            "query_id": qid,
            "chunk_id": cid,
            "question_sha256": hashlib.sha256(
                LAB["questions"][qid].encode("utf-8")
            ).hexdigest(),
            "passage_sha256": hashlib.sha256(
                chunks[cid]["content"].encode("utf-8")
            ).hexdigest(),
        }

    def c019_cached_score(identity):
        path = Path(cache_dir) / (c017_hash(identity) + ".json")
        if not path.is_file():
            return None
        record = json.loads(path.read_text(encoding="utf-8"))
        body = record["body"]
        if (
            record["body_sha256"] != c017_hash(body)
            or body["identity"] != identity
        ):
            raise RuntimeError(f"کش نامعتبر: {path}")
        value = float(body["score"])
        if not math.isfinite(value):
            raise RuntimeError(f"امتیاز کش نامعتبر: {path}")
        return value

    # Verify all prior Top20 scores before starting new inference.
    print("بررسی کش Top20 و ثابت‌بودن ورودی‌ها…", flush=True)
    for qid in ids40:
        if baseline_orders[qid] != list(LAB["incumbent"][qid]):
            raise RuntimeError(f"baseline تغییر کرده است: {qid}")
        if (
            LAB["questions"][qid]
            != golden[qid]["evaluation_content"]["question"]
        ):
            raise RuntimeError(f"متن سؤال تغییر کرده است: {qid}")
        if len(baseline_orders[qid]) < 40:
            raise RuntimeError(f"کمتر از ۴۰ نتیجه: {qid}")

        for cid in baseline_orders[qid][:20]:
            if c019_cached_score(c019_pair_identity(qid, cid)) is None:
                raise RuntimeError(
                    f"کش Top20 ناقص است؛ اجرای جدید شروع نشد: {qid}/{cid}"
                )

    print("PASS: هر ۱۲۰۰ امتیاز Top20 در کش موجود است.", flush=True)

    started40 = time.perf_counter()
    new40 = 0
    reused40 = 0
    orders40 = {}
    metrics40 = {}

    for query_index, qid in enumerate(ids40, start=1):
        candidates = baseline_orders[qid][:40]
        scores40 = {}

        for cid in candidates:
            identity = c019_pair_identity(qid, cid)
            score = c019_cached_score(identity)

            if score is None:
                encoded = tokenizer40(
                    LAB["questions"][qid],
                    chunks[cid]["content"],
                    return_tensors="pt",
                    truncation=True,
                    max_length=512,
                    padding=False,
                )
                with torch.inference_mode():
                    logits = model40(**encoded).logits

                if logits.numel() != 1:
                    raise RuntimeError("شکل خروجی مدل نامعتبر است.")
                score = float(logits.item())
                if not math.isfinite(score):
                    raise RuntimeError("امتیاز مدل متناهی نیست.")

                body = {
                    "identity": identity,
                    "score": score,
                    "input_tokens": int(encoded["input_ids"].shape[1]),
                }
                path = Path(cache_dir) / (c017_hash(identity) + ".json")
                c017_atomic_json(path, {
                    "body": body,
                    "body_sha256": c017_hash(body),
                })
                new40 += 1
            else:
                reused40 += 1

            scores40[cid] = score

        positions = {cid: i for i, cid in enumerate(candidates)}
        reranked = sorted(
            candidates, key=lambda cid: (-scores40[cid], positions[cid])
        )
        orders40[qid] = reranked + baseline_orders[qid][40:]
        metrics40[qid] = c017_metrics(orders40[qid], relevant[qid])

        print(json.dumps({
            "queries_completed": query_index,
            "queries_total": 60,
            "new_pairs": new40,
            "cached_pairs": reused40,
            "elapsed_minutes": round(
                (time.perf_counter() - started40) / 60, 2
            ),
        }), flush=True)

    aggregate40 = c017_mean(list(metrics40.values()))
    source40 = {}

    for source in ("book", "transcript", "website"):
        source_ids = [
            q for q in ids40 if source in golden[q]["source_types"]
        ]
        if source_ids:
            source40[source] = {
                "queries": len(source_ids),
                "after": c017_mean([metrics40[q] for q in source_ids]),
            }

    C019.update({
        "status": "COMPLETE",
        "method": "BGE_RERANK_BASELINE_TOP40",
        "query_count": 60,
        "contract_sha256": C017["contract_sha256"],
        "before": C017["before"],
        "bge_top20": C017["after"],
        "after": aggregate40,
        "delta_vs_baseline_percentage_points": {
            k: 100 * (aggregate40[k] - C017["before"][k])
            for k in aggregate40
        },
        "delta_vs_top20_percentage_points": {
            k: 100 * (aggregate40[k] - C017["after"][k])
            for k in aggregate40
        },
        "per_source": source40,
        "new_model_pairs": new40,
        "cached_pairs": reused40,
        "wall_seconds": time.perf_counter() - started40,
        "aggregate_no_regression_vs_baseline": all(
            aggregate40[k] >= C017["before"][k] - 1e-12
            for k in aggregate40
        ),
        "aggregate_no_regression_vs_top20": all(
            aggregate40[k] >= C017["after"][k] - 1e-12
            for k in aggregate40
        ),
        "absolute_quality_targets_met": (
            aggregate40["recall_at_10"] >= 0.85
            and aggregate40["ndcg_at_10"] >= 0.75
        ),
        "scope": "SEEN_60_DEV_QUERIES_NOT_INDEPENDENT_ACCEPTANCE",
    })

    C019["orders_artifact"] = save(
        "C019.bge_top40_orders",
        {
            "contract_sha256": C017["contract_sha256"],
            "orders": orders40,
            "metrics_by_query": metrics40,
        },
    )
    C019["report"] = save("C019.bge_top40_report", C019.copy())

    table_rows = []
    for key, label, target in (
        ("recall_at_10", "Recall@10", 0.85),
        ("ndcg_at_10", "nDCG@10", 0.75),
        ("hit_at_10", "Hit@10", None),
    ):
        threshold = "تعیین نشده" if target is None else f"{target:.0%}"
        tick = (
            "✅" if target is not None and aggregate40[key] >= target else ""
        )
        delta = 100 * (aggregate40[key] - C017["before"][key])
        table_rows.append(
            f"<tr><td>{label}</td>"
            f"<td>{C017['before'][key]:.2%}</td>"
            f"<td>{C017['after'][key]:.2%}</td>"
            f"<td>{aggregate40[key]:.2%}</td>"
            f"<td>{delta:+.2f}</td>"
            f"<td>{threshold}</td><td>{tick}</td></tr>"
        )

    display(HTML(
        '<div dir="rtl" style="text-align:right">'
        "<p>همان ۶۰ سؤال دیده‌شده؛ پذیرش مستقل نیست. "
        "تیک فقط رسیدن به حد عددی همان معیار است.</p>"
        '<table style="text-align:center">'
        "<tr><th>معیار</th><th>قبل: baseline</th>"
        "<th>BGE Top20</th><th>بعد: BGE Top40</th>"
        "<th>تغییر نسبت به baseline؛ واحد درصد</th>"
        "<th>حد پذیرش</th><th>قبول</th></tr>"
        + "".join(table_rows) + "</table></div>"
    ))
    print(json.dumps(C019, ensure_ascii=False, indent=2))

except KeyboardInterrupt:
    C019.update({
        "status": "INTERRUPTED_RESUMABLE",
        "message": "کش حفظ شده؛ برای ادامه همین سل را دوباره اجرا کن.",
    })
    print(json.dumps(C019, ensure_ascii=False, indent=2))
    raise

except Exception as exc:
    C019.update({"status": "FAILED", "error": str(exc)})
    print(json.dumps(C019, ensure_ascii=False, indent=2))
    raise
```

<li>این جداسازی در سطح سؤال است؛ استقلال منابع یا پذیرش ریلیز را ثابت نمی‌کند.</li>
<li>بدون اجرای مدل، دانلود، API یا تغییر مجموعهٔ فعال LAB.</li>

```python
# C020 v2 — Freeze BGE Top40 and reserve a new Dev cohort.
# Uses the confirmed structure: LAB["dev"]["ids"/"chunks"/"gold"].
# Does not change LAB or run/download any model.

import json
import hashlib
import unicodedata
from collections.abc import Mapping

C020 = {
    "version": "NB22E_BGE_TOP40_FREEZE_AND_COHORT_V2",
    "status": "PREPARING",
    "new_model_pairs": 0,
    "new_api_calls": 0,
    "downloads": False,
    "automatic_replacement": False,
    "release_acceptance_granted": False,
}


def c020_question_key(text):
    if not isinstance(text, str) or not text.strip():
        raise RuntimeError("متن سؤال خالی یا نامعتبر است.")
    # Conservative exact-text normalization; not a semantic duplicate test.
    return " ".join(unicodedata.normalize("NFKC", text).split())


def c020_collect_known_ids(value, known_ids, output, visited):
    """Collect Golden query IDs from existing experiment state."""
    if isinstance(value, str):
        if value in known_ids:
            output.add(value)
        return

    if not isinstance(value, (Mapping, list, tuple, set, frozenset)):
        return

    identity = id(value)
    if identity in visited:
        return
    visited.add(identity)

    if isinstance(value, Mapping):
        for key, item in value.items():
            c020_collect_known_ids(key, known_ids, output, visited)
            c020_collect_known_ids(item, known_ids, output, visited)
    else:
        for item in value:
            c020_collect_known_ids(item, known_ids, output, visited)


try:
    # ---------- Confirm the selected method ----------
    if C019.get("status") != "COMPLETE":
        raise RuntimeError("C019 کامل نیست.")
    if not C019.get("absolute_quality_targets_met"):
        raise RuntimeError("حدود عددی C019 تأیید نشده‌اند.")
    if not C019.get("aggregate_no_regression_vs_baseline"):
        raise RuntimeError("شرط عدم افت کلی نسبت به baseline برقرار نیست.")
    if C019["contract_sha256"] != C017["contract_sha256"]:
        raise RuntimeError("قرارداد C017 و C019 تطابق ندارد.")
    if c017_hash(C017["contract"]) != C017["contract_sha256"]:
        raise RuntimeError("محتوای قرارداد امتیازدهی تغییر کرده است.")

    # ---------- Read the confirmed Dev structure ----------
    dev_state = LAB["dev"]

    if not isinstance(dev_state, Mapping):
        raise RuntimeError("LAB['dev'] باید دیکشنری باشد.")

    for key in ("ids", "chunks", "gold"):
        if key not in dev_state:
            raise RuntimeError(f"کلید dev موجود نیست: {key}")

    raw_ids = dev_state["ids"]
    dev_chunks = dev_state["chunks"]
    dev_gold = dev_state["gold"]

    if not isinstance(raw_ids, (list, tuple)):
        raise RuntimeError("dev['ids'] باید فهرست شناسه‌ها باشد.")
    if not all(isinstance(q, str) for q in raw_ids):
        raise RuntimeError("شناسهٔ غیرمتنی در Development وجود دارد.")
    if len(raw_ids) != len(set(raw_ids)):
        raise RuntimeError("شناسهٔ تکراری در Development وجود دارد.")
    if not isinstance(dev_chunks, Mapping) or not isinstance(dev_gold, Mapping):
        raise RuntimeError("ساختار chunks یا gold معتبر نیست.")
    if not isinstance(golden, Mapping):
        raise RuntimeError("Golden تأییدشده در حافظه موجود نیست.")

    dev_ids = set(raw_ids)
    known_ids = set(golden)

    missing_golden = dev_ids - known_ids
    missing_gold = dev_ids - set(dev_gold)

    if missing_golden:
        raise RuntimeError(
            f"{len(missing_golden)} شناسهٔ Development در Golden نیست."
        )
    if missing_gold:
        raise RuntimeError(
            f"{len(missing_gold)} شناسهٔ Development شاهد ثبت‌شده ندارد."
        )

    # ---------- Exclude prior experiment questions ----------
    excluded = set()
    visited = set()

    # Do NOT scan LAB["dev"]: that contains the whole Dev population.
    for key in (
        "pilot", "questions", "pools", "incumbent",
        "results", "sampling", "plan",
    ):
        c020_collect_known_ids(
            LAB.get(key), known_ids, excluded, visited
        )

    for name in (
        "EXPAND60_ORIGIN",
        "C017",
        "C018",
        "C019",
    ):
        c020_collect_known_ids(
            globals().get(name), known_ids, excluded, visited
        )

    c020_collect_known_ids(
        globals().get("qids"), known_ids, excluded, visited
    )

    if not set(LAB["pilot"]).issubset(excluded):
        raise RuntimeError("سؤال‌های آزمایش قبلی کامل کنار گذاشته نشده‌اند.")

    previous_question_keys = {
        c020_question_key(golden[q]["evaluation_content"]["question"])
        for q in excluded
    }

    # ---------- Deterministic, outcome-blind selection ----------
    selected = []
    by_source = {}
    selected_question_keys = set()
    candidate_counts = {}

    for source in ("book", "transcript", "website"):
        candidates = []

        for qid in dev_ids - excluded:
            row = golden[qid]
            source_types = row.get("source_types", [])
            if isinstance(source_types, str):
                source_types = [source_types]

            if set(source_types) != {source}:
                continue

            question_key = c020_question_key(
                row["evaluation_content"]["question"]
            )
            if question_key in previous_question_keys:
                continue

            candidates.append((qid, question_key))

        candidates.sort(key=lambda item: (
            hashlib.sha256(
                ("C020_NEW_DEV_COHORT_V1|" + item[0]).encode("utf-8")
            ).hexdigest(),
            item[0],
        ))
        candidate_counts[source] = len(candidates)

        chosen = []
        for qid, question_key in candidates:
            if question_key in selected_question_keys:
                continue

            # Validate existing evidence binding; do not choose by model score.
            gold_ids = dev_gold[qid]
            if not isinstance(gold_ids, (set, frozenset, list, tuple)):
                raise RuntimeError(f"ساختار شاهد نامعتبر: {qid}")
            if not gold_ids:
                raise RuntimeError(f"شاهد ثبت‌شده خالی است: {qid}")
            if any(cid not in dev_chunks for cid in gold_ids):
                raise RuntimeError(f"قطعهٔ شاهد در dev موجود نیست: {qid}")

            chosen.append(qid)
            selected_question_keys.add(question_key)
            if len(chosen) == 20:
                break

        if len(chosen) != 20:
            raise RuntimeError(
                f"سؤال جدیدِ غیرتکراری کافی برای {source} نیست: {len(chosen)}"
            )

        by_source[source] = chosen
        selected.extend(chosen)

    if len(set(selected)) != 60 or set(selected) & excluded:
        raise RuntimeError("جداسازی شناسه‌های سؤال برقرار نیست.")

    # ---------- Record the fixed method and reserved cohort ----------
    frozen_method = {
        "method": "BGE_RERANK_BASELINE_TOP40",
        "candidate_order": "UNCHANGED_NB22D_BASELINE_ORDER",
        "depth": 40,
        "score_contract": C017["contract"],
        "score_contract_sha256": C017["contract_sha256"],
        "tie_break": "PRESERVE_BASELINE_POSITION",
        "tail": "PRESERVE_BASELINE_ORDER_AFTER_TOP40",
        "targets": {
            "recall_at_10": 0.85,
            "ndcg_at_10": 0.75,
            "hit_at_10": None,
        },
        "aggregate_no_regression_metrics": [
            "recall_at_10", "ndcg_at_10", "hit_at_10"
        ],
        "tuning_on_next_cohort": False,
        "selection_result": C019["report"],
    }

    cohort = {
        "query_ids": selected,
        "by_source": by_source,
        "excluded_query_ids": sorted(excluded),
        "candidate_counts_before_within_cohort_text_dedup": candidate_counts,
        "selection": "DETERMINISTIC_HASH_ORDER_WITHOUT_METRIC_SELECTION",
        "split": "DEVELOPMENT_ONLY",
        "question_id_overlap": 0,
        "normalized_exact_question_text_overlap": 0,
        "semantic_duplicate_independence_verified": False,
        "source_component_independence_verified": False,
        "historical_exclusion_scope": "AVAILABLE_KERNEL_STATE_ONLY",
    }

    C020["method_artifact"] = save("C020.frozen_method", frozen_method)
    C020["cohort_artifact"] = save("C020.next_dev_cohort", cohort)

    # Separate variables; active LAB cohort remains unchanged.
    C020_SELECTED_IDS = tuple(selected)
    C020_FROZEN_METHOD = frozen_method

    C020.update({
        "status": "METHOD_FROZEN_COHORT_RESERVED",
        "development_population": len(dev_ids),
        "next_query_count": len(selected),
        "source_counts": {
            source: len(ids) for source, ids in by_source.items()
        },
        "excluded_query_count": len(excluded),
        "overlap_with_excluded_questions": 0,
        "normalized_exact_question_text_overlap": 0,
        "maximum_next_model_pairs": 2400,
        "historical_exclusion_scope": "AVAILABLE_KERNEL_STATE_ONLY",
        "source_component_independence_verified": False,
        "next_step": "VERIFY_BASELINE_RECONSTRUCTION_BEFORE_NEW_COHORT_SCORING",
    })

    C020["report"] = save("C020.freeze_and_cohort_report", C020.copy())
    print(json.dumps(C020, ensure_ascii=False, indent=2))

except Exception as exc:
    C020.update({"status": "FAILED", "error": str(exc)})
    print(json.dumps(C020, ensure_ascii=False, indent=2))
    raise
```

<li>بدون دانلود، API، تغییر LAB یا جایگزینی خودکار baseline.</li>

```python
# C021 v1 — Verified NB22D replay + frozen BGE Top40 on reserved Dev60.
# Existing LAB cohort and baseline remain unchanged.

import json
import math
import time
import hashlib
from pathlib import Path
import torch
from IPython.display import display, HTML

C021 = {
    "version": "NB22E_FROZEN_BGE_NEXT_DEV60_V1",
    "status": "PREPARING",
    "downloads": False,
    "new_api_calls": 0,
    "automatic_replacement": False,
    "release_acceptance_granted": False,
}

try:
    require(
        C020.get("status") == "METHOD_FROZEN_COHORT_RESERVED",
        "C020_REQUIRED",
    )

    # Read the saved plan rather than silently reselecting questions.
    cohort21 = read_json(C020["cohort_artifact"])
    method21 = read_json(C020["method_artifact"])
    ids21 = list(cohort21["query_ids"])

    require(
        len(ids21) == len(set(ids21)) == 60,
        "COHORT_MUST_HAVE_60_UNIQUE_QUESTIONS",
    )
    require(
        not set(ids21).intersection(cohort21["excluded_query_ids"]),
        "PRIOR_QUESTION_OVERLAP",
    )
    require(
        method21["depth"] == 40
        and method21["score_contract_sha256"] == C017["contract_sha256"]
        and c017_hash(method21["score_contract"])
        == C017["contract_sha256"],
        "FROZEN_METHOD_CONFLICT",
    )

    dev21 = LAB["dev"]
    require(set(ids21).issubset(dev21["ids"]), "DEVELOPMENT_ONLY_REQUIRED")

    runtime21 = BGE_RUNTIME
    spec21 = method21["score_contract"]
    require(
        runtime21["model_id"] == spec21["model_id"]
        and runtime21["revision"] == spec21["revision"]
        and runtime21["device"] == "cpu"
        and runtime21["dtype"] == "float32"
        and torch.__version__ == spec21["torch"],
        "RUNTIME_CONTRACT_CONFLICT",
    )

    # ---------- Verify the original NB22D ranking artifact ----------
    print("بررسی فایل اصلی baseline و بازخوانی ۶۰ سؤال قبلی…", flush=True)
    d_report21 = read_json(PINS["D_final"])
    ranking_binding21 = d_report21["retained_rankings"]
    ranking_path21 = path_for(ranking_binding21["path"])

    needed21 = set(ids21) | set(LAB["pilot"])
    ranking_rows21 = {}
    h21 = hashlib.sha256()
    stat_before21 = ranking_path21.stat()

    with ranking_path21.open("rb") as stream:
        for raw in stream:
            h21.update(raw)
            row = decode(raw)
            qid = row.get("candidate_id")
            if qid in needed21:
                require(qid not in ranking_rows21, "DUPLICATE_BASELINE_QUERY")
                ranking_rows21[qid] = row

    stat_after21 = ranking_path21.stat()
    require(
        (
            stat_before21.st_size,
            stat_before21.st_mtime_ns,
            stat_before21.st_ino,
        ) == (
            stat_after21.st_size,
            stat_after21.st_mtime_ns,
            stat_after21.st_ino,
        )
        and h21.hexdigest() == binding_hash(ranking_binding21),
        "BASELINE_FILE_HASH_OR_MUTATION_CONFLICT",
    )
    require(set(ranking_rows21) == needed21, "BASELINE_QUERY_COVERAGE_CONFLICT")

    for qid in LAB["pilot"]:
        require(
            ranking_rows21[qid]["ranked_chunk_ids"]
            == LAB["incumbent"][qid],
            "PREVIOUS_BASELINE_ORDER_REPLAY_CONFLICT:" + qid,
        )

    orders_before21 = {}
    metrics_before21 = {}
    questions21 = {}

    for qid in ids21:
        row = ranking_rows21[qid]
        order = row["ranked_chunk_ids"]
        require(row["split"] == "development", "BASELINE_SPLIT_CONFLICT")
        require(
            len(order) >= 40
            and len(order) == len(set(order))
            and set(order).issubset(dev21["chunks"]),
            "INVALID_BASELINE_ORDER:" + qid,
        )

        measured = metrics(order, dev21["gold"][qid])
        require(
            all(abs(v - row["metrics"][k]) <= 1e-12
                for k, v in measured.items()),
            "SAVED_BASELINE_METRIC_REPLAY_CONFLICT:" + qid,
        )

        orders_before21[qid] = list(order)
        metrics_before21[qid] = c017_metrics(order, dev21["gold"][qid])
        questions21[qid] = golden[qid]["evaluation_content"]["question"]
        require(
            isinstance(questions21[qid], str) and questions21[qid].strip(),
            "INVALID_QUESTION_TEXT:" + qid,
        )

    # Disclose source-component dependence; do not claim a new holdout.
    component_map21 = components(dev21)
    prior_dev21 = (
        set(cohort21["excluded_query_ids"]) & set(dev21["ids"])
    )
    prior_components21 = {component_map21[q] for q in prior_dev21}
    shared_components21 = sum(
        component_map21[q] in prior_components21 for q in ids21
    )

    print("PASS: هش فایل، ترتیب قبلی و معیارهای baseline تطابق دارند.", flush=True)

    # Same pair-score identity and cache used in C017/C019.
    score_contract_hash21 = C017["contract_sha256"]
    cache_folder21 = path_for(
        "artifacts/phase2/quality_improvement/NB22E.bge_top20_cache_v1/"
        + score_contract_hash21
    )
    cache_folder21.mkdir(parents=True, exist_ok=True)

    C021["plan_artifact"] = save("C021.execution_plan", {
        "cohort": C020["cohort_artifact"],
        "frozen_method": C020["method_artifact"],
        "baseline": ranking_binding21,
        "maximum_pairs": 2400,
        "queries_sharing_source_components_with_previous": shared_components21,
        "scope": "ADDITIONAL_DEV_QUESTIONS_NOT_INDEPENDENT_HOLDOUT",
    })

    # ---------- Frozen local inference ----------
    model21 = runtime21["model"]
    tokenizer21 = runtime21["tokenizer"]
    model21.eval()
    torch.set_num_threads(spec21["threads"])

    started21 = time.perf_counter()
    new_pairs21 = 0
    cached_pairs21 = 0
    orders_after21 = {}
    metrics_after21 = {}

    for index21, qid in enumerate(ids21, start=1):
        candidates21 = orders_before21[qid][:40]
        scores21 = {}

        for cid in candidates21:
            passage = dev21["chunks"][cid]["content"]
            require(
                isinstance(passage, str) and passage.strip(),
                "INVALID_PASSAGE:" + cid,
            )
            identity = {
                "contract_sha256": score_contract_hash21,
                "query_id": qid,
                "chunk_id": cid,
                "question_sha256": hashlib.sha256(
                    questions21[qid].encode("utf-8")
                ).hexdigest(),
                "passage_sha256": hashlib.sha256(
                    passage.encode("utf-8")
                ).hexdigest(),
            }
            cache_path = cache_folder21 / (c017_hash(identity) + ".json")

            if cache_path.is_file():
                record = decode(cache_path.read_bytes())
                body = record["body"]
                require(
                    record["body_sha256"] == c017_hash(body)
                    and body["identity"] == identity,
                    "PAIR_CACHE_CONFLICT",
                )
                score = float(body["score"])
                cached_pairs21 += 1
            else:
                encoded = tokenizer21(
                    questions21[qid],
                    passage,
                    return_tensors="pt",
                    truncation=True,
                    max_length=512,
                    padding=False,
                )
                with torch.inference_mode():
                    logits = model21(**encoded).logits

                require(logits.numel() == 1, "INVALID_MODEL_OUTPUT_SHAPE")
                score = float(logits.item())
                require(math.isfinite(score), "NONFINITE_MODEL_SCORE")

                body = {
                    "identity": identity,
                    "score": score,
                    "input_tokens": int(encoded["input_ids"].shape[1]),
                }
                c017_atomic_json(cache_path, {
                    "body": body,
                    "body_sha256": c017_hash(body),
                })
                new_pairs21 += 1

            require(math.isfinite(score), "NONFINITE_CACHED_SCORE")
            scores21[cid] = score

        positions21 = {cid: i for i, cid in enumerate(candidates21)}
        ranked21 = sorted(
            candidates21,
            key=lambda cid: (-scores21[cid], positions21[cid]),
        )
        orders_after21[qid] = ranked21 + orders_before21[qid][40:]
        metrics_after21[qid] = c017_metrics(
            orders_after21[qid], dev21["gold"][qid]
        )

        print(json.dumps({
            "queries_completed": index21,
            "queries_total": 60,
            "new_pairs": new_pairs21,
            "cached_pairs": cached_pairs21,
            "elapsed_minutes": round(
                (time.perf_counter() - started21) / 60, 2
            ),
        }), flush=True)

    before21 = c017_mean(list(metrics_before21.values()))
    after21 = c017_mean(list(metrics_after21.values()))
    source_results21 = {}

    for source in ("book", "transcript", "website"):
        source_ids = cohort21["by_source"][source]
        source_results21[source] = {
            "queries": len(source_ids),
            "before": c017_mean([metrics_before21[q] for q in source_ids]),
            "after": c017_mean([metrics_after21[q] for q in source_ids]),
        }

    C021.update({
        "status": "COMPLETE",
        "query_count": 60,
        "method": "BGE_RERANK_BASELINE_TOP40",
        "contract_sha256": score_contract_hash21,
        "before": before21,
        "after": after21,
        "delta_percentage_points": {
            k: 100 * (after21[k] - before21[k]) for k in before21
        },
        "per_source": source_results21,
        "aggregate_no_regression": all(
            after21[k] >= before21[k] - 1e-12 for k in before21
        ),
        "absolute_quality_targets_met": (
            after21["recall_at_10"] >= 0.85
            and after21["ndcg_at_10"] >= 0.75
        ),
        "queries_sharing_source_components_with_previous": shared_components21,
        "new_model_pairs": new_pairs21,
        "cached_pairs": cached_pairs21,
        "wall_seconds": time.perf_counter() - started21,
        "scope": "ADDITIONAL_DEV_QUESTIONS_NOT_INDEPENDENT_HOLDOUT",
        "tuning_performed": False,
    })

    C021["orders_artifact"] = save("C021.next_dev60_orders", {
        "query_ids": ids21,
        "before_orders": orders_before21,
        "after_orders": orders_after21,
        "before_metrics": metrics_before21,
        "after_metrics": metrics_after21,
    })
    C021["report"] = save("C021.next_dev60_report", C021.copy())

    html_rows21 = []
    for key, label, target in (
        ("recall_at_10", "Recall@10", 0.85),
        ("ndcg_at_10", "nDCG@10", 0.75),
        ("hit_at_10", "Hit@10", None),
    ):
        threshold = "تعیین نشده" if target is None else f"{target:.0%}"
        tick = "✅" if target is not None and after21[key] >= target else ""
        html_rows21.append(
            f"<tr><td>{label}</td><td>{before21[key]:.2%}</td>"
            f"<td>{after21[key]:.2%}</td>"
            f"<td>{100*(after21[key]-before21[key]):+.2f}</td>"
            f"<td>{threshold}</td><td>{tick}</td></tr>"
        )

    display(HTML(
        '<div dir="rtl" style="text-align:right">'
        "<p>۶۰ سؤال دیگر از Development؛ پذیرش مستقل ریلیز نیست. "
        "تیک فقط رسیدن به حد عددی همان معیار است.</p>"
        '<table style="text-align:center">'
        "<tr><th>معیار</th><th>قبل</th><th>بعد</th>"
        "<th>تغییر؛ واحد درصد</th><th>حد پذیرش</th><th>قبول</th></tr>"
        + "".join(html_rows21) + "</table></div>"
    ))
    print(json.dumps(C021, ensure_ascii=False, indent=2))

except KeyboardInterrupt:
    C021.update({
        "status": "INTERRUPTED_RESUMABLE",
        "message": "کش حفظ شده؛ برای ادامه همین سل را دوباره اجرا کن.",
    })
    print(json.dumps(C021, ensure_ascii=False, indent=2))
    raise

except Exception as exc:
    C021.update({"status": "FAILED", "error": str(exc)})
    print(json.dumps(C021, ensure_ascii=False, indent=2))
    raise
```

```python
# C022 v2 — Correct offline audit of both BGE Top40 cohorts.
# BEFORE orders: pinned NB22D artifact.
# AFTER orders: saved C019 / C021 artifacts.
# No model inference, downloads or baseline replacement.

import json
import math
from IPython.display import display, HTML

C022 = {
    "version": "NB22E_BGE_TWO_COHORT_AUDIT_V2",
    "status": "PREPARING",
    "new_model_pairs": 0,
    "new_api_calls": 0,
    "downloads": False,
    "automatic_replacement": False,
    "release_acceptance_granted": False,
}

KEYS22 = ("recall_at_10", "ndcg_at_10", "hit_at_10")
EPS22 = 1e-12


def c022_mean(values):
    rows = list(values)
    require(bool(rows), "EMPTY_COHORT")
    return {
        key: math.fsum(row[key] for row in rows) / len(rows)
        for key in KEYS22
    }


def c022_check_metrics(actual, expected, label):
    for key in KEYS22:
        require(
            math.isfinite(actual[key])
            and math.isfinite(expected[key])
            and abs(actual[key] - expected[key]) <= 1e-9,
            "REPORT_METRIC_MISMATCH:" + label + "/" + key,
        )


def c022_summarize(ids, before, after):
    b = c022_mean(before[q] for q in ids)
    a = c022_mean(after[q] for q in ids)
    changes = {}

    for key in KEYS22:
        deltas = [after[q][key] - before[q][key] for q in ids]
        gains = [d for d in deltas if d > EPS22]
        losses = [d for d in deltas if d < -EPS22]
        positive_total = math.fsum(gains)

        changes[key] = {
            "improved_queries": len(gains),
            "worsened_queries": len(losses),
            "unchanged_queries": sum(abs(d) <= EPS22 for d in deltas),
            "largest_gain_percentage_points": 100 * max(gains, default=0.0),
            "largest_loss_percentage_points": 100 * min(losses, default=0.0),
            "largest_query_share_of_positive_gain": (
                max(gains) / positive_total if gains else None
            ),
        }

    return {
        "queries": len(ids),
        "before": b,
        "after": a,
        "delta_percentage_points": {
            key: (
                0.0 if abs(a[key] - b[key]) <= EPS22
                else 100 * (a[key] - b[key])
            )
            for key in KEYS22
        },
        "query_changes": changes,
        "aggregate_no_regression": all(
            a[key] >= b[key] - EPS22 for key in KEYS22
        ),
        "absolute_quality_targets_met": (
            a["recall_at_10"] >= 0.85
            and a["ndcg_at_10"] >= 0.75
        ),
    }


try:
    require(
        C019.get("status") == C021.get("status") == "COMPLETE",
        "BOTH_TOP40_RUNS_MUST_BE_COMPLETE",
    )

    print("خواندن و بررسی هش گزارش‌ها و ترتیب‌های ذخیره‌شده…", flush=True)

    report_a22 = read_json(C019["report"])
    report_b22 = read_json(C021["report"])

    # C017's orders are AFTER Top20 reranking.
    # Only its BEFORE metric records are used below.
    c017_saved22 = read_json(C017["orders_artifact"])
    first_bge22 = read_json(C019["orders_artifact"])
    second22 = read_json(C021["orders_artifact"])

    cohort22 = read_json(C020["cohort_artifact"])
    frozen22 = read_json(C020["method_artifact"])

    expected_contract22 = frozen22["score_contract_sha256"]
    require(
        report_a22["contract_sha256"]
        == report_b22["contract_sha256"]
        == expected_contract22,
        "MODEL_CONTRACT_MISMATCH",
    )
    require(
        c017_hash(frozen22["score_contract"]) == expected_contract22,
        "FROZEN_SCORE_CONTRACT_CHECKSUM_MISMATCH",
    )
    require(
        frozen22["depth"] == 40
        and frozen22["method"] == "BGE_RERANK_BASELINE_TOP40",
        "FROZEN_METHOD_MISMATCH",
    )

    ids_a22 = sorted(first_bge22["orders"])
    ids_b22 = list(second22["query_ids"])

    require(
        len(ids_a22) == len(set(ids_a22)) == 60
        and len(ids_b22) == len(set(ids_b22)) == 60,
        "INVALID_COHORT_SIZES",
    )
    require(
        not set(ids_a22).intersection(ids_b22),
        "QUESTION_ID_OVERLAP",
    )
    require(
        set(ids_a22) == set(c017_saved22["before_by_query"]),
        "FIRST_COHORT_METRIC_COVERAGE_MISMATCH",
    )
    require(
        set(ids_a22) == set(first_bge22["metrics_by_query"]),
        "FIRST_AFTER_METRIC_COVERAGE_MISMATCH",
    )
    require(
        set(ids_b22) == set(cohort22["query_ids"]),
        "RESERVED_COHORT_MISMATCH",
    )
    require(
        set(second22["before_orders"])
        == set(second22["after_orders"])
        == set(second22["before_metrics"])
        == set(second22["after_metrics"])
        == set(ids_b22),
        "SECOND_COHORT_COVERAGE_MISMATCH",
    )

    # ---------- Read true baseline orders from pinned NB22D ----------
    print("خواندن ترتیب اصلی baseline از خروجی NB22D…", flush=True)

    d_final22 = read_json(PINS["D_final"])
    baseline_binding22 = d_final22["retained_rankings"]
    baseline_path22 = path_for(baseline_binding22["path"])

    needed22 = set(ids_a22) | set(ids_b22)
    baseline_rows22 = {}
    baseline_hash22 = hashlib.sha256()
    stat_before22 = baseline_path22.stat()

    with baseline_path22.open("rb") as stream22:
        for raw22 in stream22:
            baseline_hash22.update(raw22)
            row22 = decode(raw22)
            q22 = row22.get("candidate_id")

            if q22 in needed22:
                require(
                    q22 not in baseline_rows22,
                    "DUPLICATE_BASELINE_QUERY:" + q22,
                )
                baseline_rows22[q22] = row22

    stat_after22 = baseline_path22.stat()
    require(
        (
            stat_before22.st_size,
            stat_before22.st_mtime_ns,
            stat_before22.st_ino,
        ) == (
            stat_after22.st_size,
            stat_after22.st_mtime_ns,
            stat_after22.st_ino,
        )
        and baseline_hash22.hexdigest() == binding_hash(baseline_binding22),
        "BASELINE_FILE_HASH_OR_MUTATION_CONFLICT",
    )
    require(
        set(baseline_rows22) == needed22,
        "BASELINE_QUERY_COVERAGE_CONFLICT",
    )

    before_orders22 = {}

    for q22 in sorted(needed22):
        row22 = baseline_rows22[q22]
        order22 = row22["ranked_chunk_ids"]

        require(
            row22["split"] == "development",
            "BASELINE_SPLIT_CONFLICT:" + q22,
        )
        require(
            len(order22) >= 40
            and len(order22) == len(set(order22))
            and set(order22).issubset(LAB["dev"]["chunks"]),
            "INVALID_BASELINE_ORDER:" + q22,
        )

        replay22 = metrics(order22, LAB["dev"]["gold"][q22])
        require(
            all(
                abs(value22 - row22["metrics"][key22]) <= 1e-12
                for key22, value22 in replay22.items()
            ),
            "NB22D_BASELINE_REPLAY_CONFLICT:" + q22,
        )
        before_orders22[q22] = order22

    # C021 saved explicit BEFORE orders; confirm their provenance too.
    for q22 in ids_b22:
        require(
            second22["before_orders"][q22] == before_orders22[q22],
            "SECOND_SAVED_BASELINE_ORDER_MISMATCH:" + q22,
        )

    # ---------- Recompute every query's metrics ----------
    before22 = {}
    after22 = {}
    query_rows22 = []

    for cohort_name22, cohort_ids22, after_orders22, saved_before22, saved_after22 in (
        (
            "selection_dev60",
            ids_a22,
            first_bge22["orders"],
            c017_saved22["before_by_query"],
            first_bge22["metrics_by_query"],
        ),
        (
            "additional_dev60",
            ids_b22,
            second22["after_orders"],
            second22["before_metrics"],
            second22["after_metrics"],
        ),
    ):
        for q22 in cohort_ids22:
            gold22 = LAB["dev"]["gold"][q22]
            b_order22 = before_orders22[q22]
            a_order22 = after_orders22[q22]

            require(
                len(a_order22) == len(b_order22)
                and len(a_order22) == len(set(a_order22))
                and set(a_order22[:40]) == set(b_order22[:40])
                and a_order22[40:] == b_order22[40:],
                "TOP40_RERANK_CONTRACT_MISMATCH:" + q22,
            )

            b_all22 = metrics(b_order22, gold22)
            a_all22 = metrics(a_order22, gold22)
            b22 = {key: b_all22[key] for key in KEYS22}
            a22 = {key: a_all22[key] for key in KEYS22}

            c022_check_metrics(b22, saved_before22[q22], "before/" + q22)
            c022_check_metrics(a22, saved_after22[q22], "after/" + q22)

            before22[q22] = b22
            after22[q22] = a22
            query_rows22.append({
                "query_id": q22,
                "cohort": cohort_name22,
                "source_types": golden[q22]["source_types"],
                "before": b22,
                "after": a22,
                "delta_percentage_points": {
                    key: (
                        0.0 if abs(a22[key] - b22[key]) <= EPS22
                        else 100 * (a22[key] - b22[key])
                    )
                    for key in KEYS22
                },
            })

    summary_a22 = c022_summarize(ids_a22, before22, after22)
    summary_b22 = c022_summarize(ids_b22, before22, after22)

    c022_check_metrics(
        summary_a22["before"], report_a22["before"], "cohort_A_before"
    )
    c022_check_metrics(
        summary_a22["after"], report_a22["after"], "cohort_A_after"
    )
    c022_check_metrics(
        summary_b22["before"], report_b22["before"], "cohort_B_before"
    )
    c022_check_metrics(
        summary_b22["after"], report_b22["after"], "cohort_B_after"
    )

    all_ids22 = ids_a22 + ids_b22
    combined22 = c022_summarize(all_ids22, before22, after22)

    source_summary22 = {}
    for cohort_name22, cohort_ids22 in (
        ("selection_dev60", ids_a22),
        ("additional_dev60", ids_b22),
    ):
        source_summary22[cohort_name22] = {}

        for source22 in ("book", "transcript", "website"):
            source_ids22 = [
                q for q in cohort_ids22
                if source22 in golden[q]["source_types"]
            ]
            if source_ids22:
                source_summary22[cohort_name22][source22] = (
                    c022_summarize(source_ids22, before22, after22)
                )

    second_passed22 = (
        summary_b22["aggregate_no_regression"]
        and summary_b22["absolute_quality_targets_met"]
    )

    C022.update({
        "status": "COMPLETE",
        "baseline_binding": baseline_binding22,
        "contract_sha256": expected_contract22,
        "cohorts": {
            "selection_dev60": summary_a22,
            "additional_dev60": summary_b22,
        },
        "combined_descriptive_only": combined22,
        "per_source": source_summary22,
        "question_id_overlap": 0,
        "all_saved_metrics_reproduced": True,
        "all_top40_order_contracts_verified": True,
        "additional_cohort_point_conditions_met": second_passed22,
        "decision": (
            "ADDITIONAL_DEV_POINT_CONDITIONS_MET_NOT_RELEASE_ACCEPTANCE"
            if second_passed22
            else "KEEP_BASELINE_NO_AUTOMATIC_REPLACEMENT"
        ),
        "scope": "DEVELOPMENT_ONLY_NOT_INDEPENDENT_RELEASE_ACCEPTANCE",
    })

    C022["query_changes_artifact"] = save(
        "C022.query_changes", query_rows22, rows=True
    )
    C022["report"] = save("C022.two_cohort_report", C022.copy())

    # ---------- Tables ----------
    table_rows22 = []

    for cohort_label22, result22 in (
        ("۶۰ سؤال اول؛ انتخاب روش", summary_a22),
        ("۶۰ سؤال دوم؛ روش ثابت", summary_b22),
        ("۱۲۰ سؤال؛ صرفاً توصیفی", combined22),
    ):
        for key22, label22, target22 in (
            ("recall_at_10", "Recall@10", 0.85),
            ("ndcg_at_10", "nDCG@10", 0.75),
            ("hit_at_10", "Hit@10", None),
        ):
            threshold22 = (
                "تعیین نشده" if target22 is None else f"{target22:.0%}"
            )
            tick22 = (
                "✅"
                if target22 is not None
                and result22["after"][key22] >= target22
                else ""
            )
            table_rows22.append(
                f"<tr><td>{cohort_label22}</td><td>{label22}</td>"
                f"<td>{result22['before'][key22]:.2%}</td>"
                f"<td>{result22['after'][key22]:.2%}</td>"
                f"<td>{result22['delta_percentage_points'][key22]:+.2f}</td>"
                f"<td>{threshold22}</td><td>{tick22}</td></tr>"
            )

    change_rows22 = []

    for cohort_label22, result22 in (
        ("۶۰ سؤال اول", summary_a22),
        ("۶۰ سؤال دوم", summary_b22),
    ):
        for key22, label22 in (
            ("recall_at_10", "Recall@10"),
            ("ndcg_at_10", "nDCG@10"),
            ("hit_at_10", "Hit@10"),
        ):
            changes22 = result22["query_changes"][key22]
            change_rows22.append(
                f"<tr><td>{cohort_label22}</td><td>{label22}</td>"
                f"<td>{changes22['improved_queries']}</td>"
                f"<td>{changes22['worsened_queries']}</td>"
                f"<td>{changes22['unchanged_queries']}</td></tr>"
            )

    display(HTML(
        '<div dir="rtl" style="text-align:right">'
        "<p><strong>بازمحاسبه و تطبیق گزارش‌ها موفق بود.</strong> "
        "تیک فقط حد عددی همان معیار است؛ پذیرش ریلیز نیست.</p>"
        '<table style="text-align:center">'
        "<tr><th>مجموعه</th><th>معیار</th><th>قبل</th><th>بعد</th>"
        "<th>تغییر؛ واحد درصد</th><th>حد پذیرش</th><th>قبول</th></tr>"
        + "".join(table_rows22)
        + "</table><br>"
        '<table style="text-align:center">'
        "<tr><th>مجموعه</th><th>معیار</th>"
        "<th>سؤال بهتر</th><th>سؤال بدتر</th><th>بدون تغییر</th></tr>"
        + "".join(change_rows22)
        + "</table></div>"
    ))

    print(json.dumps(C022, ensure_ascii=False, indent=2))

except Exception as exc:
    C022.update({"status": "FAILED", "error": str(exc)})
    print(json.dumps(C022, ensure_ascii=False, indent=2))
    raise
```

<h3>C023 — نسخهٔ ۱: ترکیب ثابت رتبه‌های baseline و BGE</h3>

<li>هر دو مجموعه اکنون دادهٔ توسعه‌اند؛ این آزمایش پذیرش مستقل نیست.</li>

```python
# C023 v1 — One fixed, rank-only fusion experiment.
# Equal-weight RRF, k=60. No model execution or parameter search.

import json
import math
from IPython.display import display, HTML

C023 = {
    "version": "NB22E_FIXED_BGE_BASELINE_RRF_V1",
    "status": "PREPARING",
    "new_model_pairs": 0,
    "new_api_calls": 0,
    "downloads": False,
    "automatic_replacement": False,
    "release_acceptance_granted": False,
}

try:
    require(C022.get("status") == "COMPLETE", "C022_REQUIRED")

    audit23 = read_json(C022["report"])
    require(
        audit23["all_saved_metrics_reproduced"]
        and audit23["all_top40_order_contracts_verified"],
        "VERIFIED_TWO_COHORT_AUDIT_REQUIRED",
    )

    changes23 = read_rows(C022["query_changes_artifact"])
    first23 = read_json(C019["orders_artifact"])
    second23 = read_json(C021["orders_artifact"])

    cohorts23 = {
        "selection_dev60": sorted(first23["orders"]),
        "additional_dev60": list(second23["query_ids"]),
    }
    all_ids23 = (
        cohorts23["selection_dev60"] + cohorts23["additional_dev60"]
    )
    require(
        len(all_ids23) == len(set(all_ids23)) == 120,
        "COHORT_SIZE_OR_OVERLAP_CONFLICT",
    )

    # Expected per-query metrics from the completed audit.
    expected23 = {}
    for row in changes23:
        qid = row["query_id"]
        require(qid not in expected23, "DUPLICATE_AUDIT_QUERY")
        expected23[qid] = row
    require(set(expected23) == set(all_ids23), "AUDIT_COVERAGE_CONFLICT")

    # Read the true baseline, never the reranked C017 orders.
    baseline_rows23 = {}
    needed23 = set(all_ids23)
    binding23 = audit23["baseline_binding"]
    path23 = path_for(binding23["path"])
    hash23 = hashlib.sha256()
    stat_before23 = path23.stat()

    print("بررسی ترتیب‌های ذخیره‌شده و baseline…", flush=True)

    with path23.open("rb") as stream:
        for raw in stream:
            hash23.update(raw)
            row = decode(raw)
            qid = row.get("candidate_id")
            if qid in needed23:
                require(qid not in baseline_rows23, "DUPLICATE_BASELINE_QUERY")
                baseline_rows23[qid] = row

    stat_after23 = path23.stat()
    require(
        (
            stat_before23.st_size,
            stat_before23.st_mtime_ns,
            stat_before23.st_ino,
        ) == (
            stat_after23.st_size,
            stat_after23.st_mtime_ns,
            stat_after23.st_ino,
        )
        and hash23.hexdigest() == binding_hash(binding23),
        "BASELINE_HASH_OR_MUTATION_CONFLICT",
    )
    require(set(baseline_rows23) == needed23, "BASELINE_COVERAGE_CONFLICT")

    # Freeze this single experiment before evaluating it.
    spec23 = {
        "method": "EQUAL_WEIGHT_RRF_BASELINE_BGE_TOP40",
        "rrf_k": 60,
        "baseline_weight": 0.5,
        "bge_weight": 0.5,
        "depth": 40,
        "tie_break": "BASELINE_POSITION",
        "tail": "UNCHANGED_BASELINE_AFTER_TOP40",
        "parameter_search": False,
        "scope": "BOTH_COHORTS_ARE_NOW_DEVELOPMENT_DATA",
    }
    C023["specification_artifact"] = save("C023.fixed_rrf_spec", spec23)

    before23 = {}
    bge23 = {}
    fusion23 = {}
    fusion_orders23 = {}
    query_changes23 = []
    metric_keys23 = ("recall_at_10", "ndcg_at_10", "hit_at_10")

    for cohort_name, ids in cohorts23.items():
        bge_orders = (
            first23["orders"]
            if cohort_name == "selection_dev60"
            else second23["after_orders"]
        )

        for qid in ids:
            require(
                expected23[qid]["cohort"] == cohort_name,
                "AUDIT_COHORT_CONFLICT",
            )
            row = baseline_rows23[qid]
            require(row["split"] == "development", "SPLIT_CONFLICT")
            baseline = row["ranked_chunk_ids"]
            bge = bge_orders[qid]

            require(
                len(baseline) >= 40
                and len(baseline) == len(set(baseline))
                and len(bge) == len(baseline)
                and len(bge) == len(set(bge))
                and set(bge[:40]) == set(baseline[:40])
                and bge[40:] == baseline[40:],
                "TOP40_ORDER_CONTRACT_CONFLICT:" + qid,
            )

            gold = LAB["dev"]["gold"][qid]
            baseline_all = metrics(baseline, gold)
            bge_all = metrics(bge, gold)
            before23[qid] = {k: baseline_all[k] for k in metric_keys23}
            bge23[qid] = {k: bge_all[k] for k in metric_keys23}

            for key in metric_keys23:
                require(
                    abs(
                        before23[qid][key]
                        - expected23[qid]["before"][key]
                    ) <= 1e-9
                    and abs(
                        bge23[qid][key]
                        - expected23[qid]["after"][key]
                    ) <= 1e-9,
                    "AUDITED_METRIC_REPLAY_CONFLICT:" + qid + "/" + key,
                )

            baseline_rank = {
                cid: i + 1 for i, cid in enumerate(baseline[:40])
            }
            bge_rank = {
                cid: i + 1 for i, cid in enumerate(bge[:40])
            }

            rrf_scores = {
                cid: (
                    0.5 / (60 + baseline_rank[cid])
                    + 0.5 / (60 + bge_rank[cid])
                )
                for cid in baseline[:40]
            }
            fused_head = sorted(
                baseline[:40],
                key=lambda cid: (-rrf_scores[cid], baseline_rank[cid]),
            )
            fused_order = fused_head + baseline[40:]
            fused_all = metrics(fused_order, gold)

            fusion_orders23[qid] = fused_order
            fusion23[qid] = {k: fused_all[k] for k in metric_keys23}

            query_changes23.append({
                "query_id": qid,
                "cohort": cohort_name,
                "before": before23[qid],
                "pure_bge": bge23[qid],
                "fusion": fusion23[qid],
            })

    # Reuse the verified summary helper from C022.
    results23 = {}
    for cohort_name, ids in cohorts23.items():
        result = c022_summarize(ids, before23, fusion23)
        result["pure_bge"] = c022_mean(bge23[q] for q in ids)
        result["passes_fixed_experiment_guard"] = (
            result["aggregate_no_regression"]
            and result["after"]["ndcg_at_10"]
            > result["before"]["ndcg_at_10"] + 1e-12
        )

        result["per_source"] = {}
        for source in ("book", "transcript", "website"):
            source_ids = [
                q for q in ids if source in golden[q]["source_types"]
            ]
            if source_ids:
                result["per_source"][source] = c022_summarize(
                    source_ids, before23, fusion23
                )
        results23[cohort_name] = result

    passes_both23 = all(
        r["passes_fixed_experiment_guard"] for r in results23.values()
    )

    C023.update({
        "status": "COMPLETE",
        "specification": spec23,
        "cohorts": results23,
        "passes_guard_in_both_cohorts": passes_both23,
        "absolute_targets_met_in_both_cohorts": all(
            r["absolute_quality_targets_met"] for r in results23.values()
        ),
        "decision": (
            "EXPLORATORY_CANDIDATE_ONLY_REQUIRES_SEPARATE_VALIDATION"
            if passes_both23
            else "FIXED_FUSION_NOT_SELECTED_KEEP_BASELINE"
        ),
        "scope": "POST_OUTCOME_DEVELOPMENT_EXPERIMENT_NOT_INDEPENDENT_ACCEPTANCE",
    })

    C023["orders_artifact"] = save("C023.fixed_rrf_orders", {
        "specification": spec23,
        "orders": fusion_orders23,
        "query_metrics": query_changes23,
    })
    C023["report"] = save("C023.fixed_rrf_report", C023.copy())

    table_rows23 = []
    for cohort_name, label in (
        ("selection_dev60", "۶۰ سؤال اول"),
        ("additional_dev60", "۶۰ سؤال دوم"),
    ):
        result = results23[cohort_name]

        for key, metric_label, target in (
            ("recall_at_10", "Recall@10", 0.85),
            ("ndcg_at_10", "nDCG@10", 0.75),
            ("hit_at_10", "Hit@10", None),
        ):
            threshold = "تعیین نشده" if target is None else f"{target:.0%}"
            tick = (
                "✅"
                if target is not None and result["after"][key] >= target
                else ""
            )
            table_rows23.append(
                f"<tr><td>{label}</td><td>{metric_label}</td>"
                f"<td>{result['before'][key]:.2%}</td>"
                f"<td>{result['pure_bge'][key]:.2%}</td>"
                f"<td>{result['after'][key]:.2%}</td>"
                f"<td>{result['delta_percentage_points'][key]:+.2f}</td>"
                f"<td>{threshold}</td><td>{tick}</td></tr>"
            )

    display(HTML(
        '<div dir="rtl" style="text-align:right">'
        "<p>یک آزمایش با فرمول ثابت؛ هر دو مجموعه دادهٔ توسعه‌اند. "
        "تیک فقط حد عددی معیار را نشان می‌دهد.</p>"
        '<table style="text-align:center">'
        "<tr><th>مجموعه</th><th>معیار</th><th>قبل: baseline</th>"
        "<th>BGE خالص</th><th>بعد: ترکیب ثابت</th>"
        "<th>تغییر نسبت به baseline؛ واحد درصد</th>"
        "<th>حد پذیرش</th><th>قبول</th></tr>"
        + "".join(table_rows23)
        + "</table></div>"
    ))

    print(json.dumps(C023, ensure_ascii=False, indent=2))

except Exception as exc:
    C023.update({"status": "FAILED", "error": str(exc)})
    print(json.dumps(C023, ensure_ascii=False, indent=2))
    raise
```

<h3>C024 — نسخهٔ ۱: سقف پوشش Top40 و فاصلهٔ رتبه‌بندی</h3>
<ul>
<li>بررسی جداگانهٔ هر دو مجموعهٔ ۶۰تایی و منابع آن‌ها.</li>
<li>سقف Recall@10: بهترین نتیجهٔ نظری با شاهدهای موجود در Top40.</li>
<li>فاصلهٔ رتبه‌بندی: اختلاف سقف نظری با نتیجهٔ ترکیب ثابت.</li>
<li>گزارش جداگانهٔ شاهدهای خارج از Top40 و محدودیت خروجی ۱۰تایی.</li>
<li>این محاسبات علت معنایی خطا یا عملکرد قطعی مدل دیگری را ثابت نمی‌کنند.</li>
<li>بدون اجرای مدل، دانلود، API یا تغییر تنظیمات.</li>
</ul>

```python
# C024 v1 — Offline Top40 coverage and ranking-gap decomposition.
# All quantities are macro averages over questions.
# No inference, downloads, tuning or automatic replacement.

import json
import math
import hashlib
from IPython.display import display, HTML

C024 = {
    "version": "NB22E_TOP40_COVERAGE_DECOMPOSITION_V1",
    "status": "PREPARING",
    "new_model_pairs": 0,
    "new_api_calls": 0,
    "downloads": False,
    "automatic_replacement": False,
    "release_acceptance_granted": False,
}

try:
    require(C023.get("status") == "COMPLETE", "C023_REQUIRED")
    require(C022.get("status") == "COMPLETE", "C022_REQUIRED")

    report24 = read_json(C023["report"])
    fusion_artifact24 = read_json(C023["orders_artifact"])
    audit24 = read_json(C022["report"])

    require(
        fusion_artifact24["specification"] == report24["specification"],
        "FUSION_SPECIFICATION_CONFLICT",
    )

    query_records24 = {}
    for row24 in fusion_artifact24["query_metrics"]:
        q24 = row24["query_id"]
        require(q24 not in query_records24, "DUPLICATE_QUERY_RECORD")
        query_records24[q24] = row24

    fused_orders24 = fusion_artifact24["orders"]
    require(
        set(query_records24) == set(fused_orders24)
        and len(query_records24) == 120,
        "EXPECTED_120_QUERY_COVERAGE",
    )

    # Verify original baseline artifact and retain only required queries.
    binding24 = audit24["baseline_binding"]
    path24 = path_for(binding24["path"])
    baseline24 = {}
    hash24 = hashlib.sha256()
    stat_before24 = path24.stat()

    print("بررسی baseline و خروجی ترکیب ثابت…", flush=True)

    with path24.open("rb") as stream24:
        for raw24 in stream24:
            hash24.update(raw24)
            row24 = decode(raw24)
            q24 = row24.get("candidate_id")

            if q24 in query_records24:
                require(q24 not in baseline24, "DUPLICATE_BASELINE_QUERY")
                require(row24["split"] == "development", "SPLIT_CONFLICT")
                baseline24[q24] = row24["ranked_chunk_ids"]

    stat_after24 = path24.stat()
    require(
        (
            stat_before24.st_size,
            stat_before24.st_mtime_ns,
            stat_before24.st_ino,
        ) == (
            stat_after24.st_size,
            stat_after24.st_mtime_ns,
            stat_after24.st_ino,
        )
        and hash24.hexdigest() == binding_hash(binding24),
        "BASELINE_HASH_OR_MUTATION_CONFLICT",
    )
    require(
        set(baseline24) == set(query_records24),
        "BASELINE_QUERY_COVERAGE_CONFLICT",
    )

    cases24 = []

    for q24 in sorted(query_records24):
        record24 = query_records24[q24]
        order24 = baseline24[q24]
        fused24 = fused_orders24[q24]
        gold24 = set(LAB["dev"]["gold"][q24])

        require(bool(gold24), "EMPTY_GOLD:" + q24)
        require(
            len(order24) >= 40
            and len(order24) == len(set(order24))
            and len(fused24) == len(order24)
            and len(fused24) == len(set(fused24))
            and set(fused24[:40]) == set(order24[:40])
            and fused24[40:] == order24[40:],
            "TOP40_ORDER_CONTRACT_CONFLICT:" + q24,
        )

        before_metrics24 = metrics(order24, gold24)
        after_metrics24 = metrics(fused24, gold24)

        for key24 in ("recall_at_10", "ndcg_at_10", "hit_at_10"):
            require(
                abs(before_metrics24[key24] - record24["before"][key24])
                <= 1e-9
                and abs(after_metrics24[key24] - record24["fusion"][key24])
                <= 1e-9,
                "SAVED_METRIC_REPLAY_CONFLICT:" + q24 + "/" + key24,
            )

        pool24 = set(order24[:40])
        available24 = gold24 & pool24
        found24 = gold24 & set(fused24[:10])
        total_gold24 = len(gold24)

        ceiling24 = min(10, len(available24)) / total_gold24
        actual24 = after_metrics24["recall_at_10"]

        outside_pool_gap24 = len(gold24 - pool24) / total_gold24
        output_capacity_gap24 = max(
            0, len(available24) - 10
        ) / total_gold24
        ranking_gap24 = ceiling24 - actual24

        require(ranking_gap24 >= -1e-12, "NEGATIVE_RANKING_GAP")
        ranking_gap24 = max(0.0, ranking_gap24)

        # Exact decomposition of the gap from Recall=1.
        require(
            abs(
                (1 - actual24)
                - outside_pool_gap24
                - output_capacity_gap24
                - ranking_gap24
            ) <= 1e-12,
            "GAP_DECOMPOSITION_CONFLICT:" + q24,
        )

        cases24.append({
            "query_id": q24,
            "cohort": record24["cohort"],
            "source_types": golden[q24]["source_types"],
            "baseline_recall_at_10": before_metrics24["recall_at_10"],
            "fusion_recall_at_10": actual24,
            "top40_recall_at_10_ceiling": ceiling24,
            "outside_top40_gap": outside_pool_gap24,
            "top10_capacity_gap": output_capacity_gap24,
            "ranking_gap": ranking_gap24,
            "gold_count": total_gold24,
            "gold_in_top40": len(available24),
            "gold_in_fusion_top10": len(found24),
            "gold_outside_top40": sorted(gold24 - pool24),
            "gold_in_top40_not_in_fusion_top10": sorted(available24 - found24),
        })

    def summarize24(rows24):
        require(bool(rows24), "EMPTY_SUMMARY_GROUP")
        keys24 = (
            "baseline_recall_at_10",
            "fusion_recall_at_10",
            "top40_recall_at_10_ceiling",
            "outside_top40_gap",
            "top10_capacity_gap",
            "ranking_gap",
        )
        result24 = {
            key24: math.fsum(r[key24] for r in rows24) / len(rows24)
            for key24 in keys24
        }
        result24.update({
            "queries": len(rows24),
            "queries_with_gold_outside_top40": sum(
                bool(r["gold_outside_top40"]) for r in rows24
            ),
            "queries_with_ranking_gap": sum(
                r["ranking_gap"] > 1e-12 for r in rows24
            ),
            "recall_target": 0.85,
            "ceiling_has_target_capacity": (
                result24["top40_recall_at_10_ceiling"] >= 0.85
            ),
            "actual_recall_target_met": (
                result24["fusion_recall_at_10"] >= 0.85
            ),
        })
        return result24

    cohorts24 = {}
    source_slices24 = {}

    for name24 in ("selection_dev60", "additional_dev60"):
        rows24 = [r for r in cases24 if r["cohort"] == name24]
        require(len(rows24) == 60, "COHORT_SIZE_CONFLICT:" + name24)
        cohorts24[name24] = summarize24(rows24)
        source_slices24[name24] = {}

        # Verify aggregate Recall against the saved C023 report.
        require(
            abs(
                cohorts24[name24]["fusion_recall_at_10"]
                - report24["cohorts"][name24]["after"]["recall_at_10"]
            ) <= 1e-9,
            "AGGREGATE_RECALL_REPLAY_CONFLICT",
        )

        for source24 in ("book", "transcript", "website"):
            source_rows24 = [
                r for r in rows24 if source24 in r["source_types"]
            ]
            if source_rows24:
                source_slices24[name24][source24] = summarize24(source_rows24)

    C024.update({
        "status": "COMPLETE",
        "cohorts": cohorts24,
        "per_source": source_slices24,
        "combined_descriptive_only": summarize24(cases24),
        "decomposition": (
            "1 - actual Recall@10 = outside Top40 gap "
            "+ Top10 capacity gap + ranking gap"
        ),
        "interpretation_limit": (
            "Oracle coverage and ranking opportunity only; "
            "not a semantic diagnosis or guaranteed achievable model quality."
        ),
        "scope": "DEVELOPMENT_ONLY_NOT_INDEPENDENT_ACCEPTANCE",
    })

    C024["cases_artifact"] = save(
        "C024.coverage_cases", cases24, rows=True
    )
    C024["report"] = save("C024.coverage_report", C024.copy())

    html_rows24 = []
    groups24 = []

    for name24, label24 in (
        ("selection_dev60", "۶۰ سؤال اول"),
        ("additional_dev60", "۶۰ سؤال دوم"),
    ):
        groups24.append((label24, cohorts24[name24]))
        for source24, source_label24 in (
            ("book", "کتاب"),
            ("transcript", "متن صوتی"),
            ("website", "وب‌سایت"),
        ):
            if source24 in source_slices24[name24]:
                groups24.append((
                    label24 + "؛ " + source_label24,
                    source_slices24[name24][source24],
                ))

    for label24, result24 in groups24:
        actual_tick24 = "✅" if result24["actual_recall_target_met"] else ""
        capacity_tick24 = (
            "✅" if result24["ceiling_has_target_capacity"] else ""
        )
        html_rows24.append(
            f"<tr><td>{label24}</td>"
            f"<td>{result24['baseline_recall_at_10']:.2%}</td>"
            f"<td>{result24['fusion_recall_at_10']:.2%}</td>"
            f"<td>{result24['top40_recall_at_10_ceiling']:.2%}</td>"
            f"<td>{100*result24['ranking_gap']:.2f}</td>"
            f"<td>{100*result24['outside_top40_gap']:.2f}</td>"
            f"<td>{100*result24['top10_capacity_gap']:.2f}</td>"
            f"<td>85%</td><td>{actual_tick24}</td>"
            f"<td>{capacity_tick24}</td></tr>"
        )

    display(HTML(
        '<div dir="rtl" style="text-align:right">'
        "<p>همهٔ اعداد مربوط به Recall@10 هستند. "
        "فاصله‌ها بر حسب واحد درصدند. "
        "سقف نظری، نتیجهٔ واقعی مدل نیست.</p>"
        '<table style="text-align:center">'
        "<tr><th>مجموعه</th><th>قبل</th><th>بعدِ ترکیب</th>"
        "<th>سقف Top40</th><th>فاصلهٔ رتبه‌بندی</th>"
        "<th>فاصلهٔ خارج Top40</th><th>محدودیت Top10</th>"
        "<th>هدف</th><th>قبول واقعی</th><th>ظرفیت نظری کافی</th></tr>"
        + "".join(html_rows24) + "</table></div>"
    ))

    print(json.dumps(C024, ensure_ascii=False, indent=2))

except Exception as exc:
    C024.update({"status": "FAILED", "error": str(exc)})
    print(json.dumps(C024, ensure_ascii=False, indent=2))
    raise
```

<div dir="rtl" style="text-align:right">
<h3>C025 — نسخهٔ ۱: مقایسهٔ پوشش Top40، Top100 و pool کامل NB22D</h3>
<ul>
<li>بررسی همان ۱۲۰ سؤال، جداگانه برای دو مجموعه و سه منبع.</li>
<li>خواندن فایل‌های اصلی baseline و pool با بررسی هش.</li>
<li>محاسبهٔ سقف نظری Recall@10 برای هر مجموعهٔ ورودی.</li>
<li>سنجش شاهدهای اضافه یا ازدست‌رفته نسبت به Top40.</li>
<li>هدف ۸۵٪، هدف کلی است؛ بررسی منابع جنبهٔ تشخیصی دارد.</li>
<li>بدون اجرای مدل، دانلود، API، تنظیم وزن یا جایگزینی baseline.</li>
<li>سقف نظری تضمین کیفیت واقعی رتبه‌بندی نیست.</li>
</ul>
</div>

```python
# C025 v1 — Offline candidate coverage across 120 Dev questions.
# Compare baseline Top40, baseline Top100, and the saved NB22D pool.
# No model execution, downloads, tuning, or automatic replacement.

import json
import math
import hashlib
from IPython.display import display, HTML

C025 = {
    "version": "NB22E_CANDIDATE_COVERAGE_COMPARISON_V1",
    "status": "PREPARING",
    "new_model_pairs": 0,
    "new_api_calls": 0,
    "downloads": False,
    "automatic_replacement": False,
    "release_acceptance_granted": False,
}


def c025_subset(binding, needed):
    path = path_for(binding["path"])
    before = path.stat()
    hasher = hashlib.sha256()
    selected = {}

    with path.open("rb") as stream:
        for raw in stream:
            hasher.update(raw)
            row = decode(raw)
            qid = row.get("candidate_id")
            if qid in needed:
                require(qid not in selected, "DUPLICATE_QUERY:" + qid)
                selected[qid] = row

    after = path.stat()
    require(
        (before.st_size, before.st_mtime_ns, before.st_ino)
        == (after.st_size, after.st_mtime_ns, after.st_ino)
        and hasher.hexdigest() == binding_hash(binding),
        "SOURCE_HASH_OR_MUTATION_CONFLICT:" + binding["path"],
    )
    require(set(selected) == needed, "SOURCE_QUERY_COVERAGE_CONFLICT")
    return selected


def c025_mean(rows):
    require(bool(rows), "EMPTY_SUMMARY")
    numeric = (
        "baseline_recall_at_10",
        "fusion_recall_at_10",
        "recall_at_10_ceiling",
        "ceiling_delta_vs_top40",
        "outside_candidate_gap",
        "top10_capacity_gap",
        "added_gold_fraction_vs_top40",
        "lost_gold_fraction_vs_top40",
    )
    result = {
        key: math.fsum(row[key] for row in rows) / len(rows)
        for key in numeric
    }
    sizes = sorted(row["candidate_count"] for row in rows)
    result.update({
        "queries": len(rows),
        "candidate_count_min": sizes[0],
        "candidate_count_max": sizes[-1],
        "candidate_count_mean": math.fsum(sizes) / len(sizes),
        "queries_with_added_gold_vs_top40": sum(
            row["added_gold_count_vs_top40"] > 0 for row in rows
        ),
        "queries_with_lost_gold_vs_top40": sum(
            row["lost_gold_count_vs_top40"] > 0 for row in rows
        ),
        "queries_with_gold_outside_candidates": sum(
            row["outside_gold_count"] > 0 for row in rows
        ),
        "recall_target": 0.85,
        "ceiling_has_target_capacity": (
            result["recall_at_10_ceiling"] >= 0.85
        ),
        "capacity_only_not_actual_acceptance": True,
    })
    return result


try:
    require(C024.get("status") == "COMPLETE", "C024_REQUIRED")

    coverage_report25 = read_json(C024["report"])
    prior_cases25 = read_rows(C024["cases_artifact"])

    prior_by_query25 = {}
    for row25 in prior_cases25:
        q25 = row25["query_id"]
        require(q25 not in prior_by_query25, "DUPLICATE_C024_QUERY")
        prior_by_query25[q25] = row25

    needed25 = set(prior_by_query25)
    require(len(needed25) == 120, "EXPECTED_120_QUERIES")

    cohort_ids25 = {
        name25: sorted(
            q for q, row in prior_by_query25.items()
            if row["cohort"] == name25
        )
        for name25 in ("selection_dev60", "additional_dev60")
    }
    require(
        all(len(ids) == 60 for ids in cohort_ids25.values())
        and set().union(*map(set, cohort_ids25.values())) == needed25,
        "COHORT_MEMBERSHIP_CONFLICT",
    )

    print("بررسی هش فایل‌های اصلی baseline و pool مربوط به NB22D…", flush=True)

    d_final25 = read_json(PINS["D_final"])
    rank_binding25 = d_final25["retained_rankings"]
    pool_binding25 = d_final25["pool_artifact"]

    ranks25 = c025_subset(rank_binding25, needed25)
    pools25 = c025_subset(pool_binding25, needed25)
    dev25 = LAB["dev"]

    cases25 = []

    for q25 in sorted(needed25):
        require(
            ranks25[q25]["split"] == pools25[q25]["split"] == "development",
            "DEVELOPMENT_SPLIT_REQUIRED:" + q25,
        )

        baseline25 = ranks25[q25]["ranked_chunk_ids"]
        pool25 = pools25[q25]["candidate_chunk_ids"]

        require(
            len(baseline25) >= 100
            and len(baseline25) == len(set(baseline25))
            and set(baseline25).issubset(dev25["chunks"]),
            "INVALID_BASELINE_ORDER:" + q25,
        )
        require(
            0 < len(pool25) <= 800
            and len(pool25) == len(set(pool25))
            and set(pool25).issubset(dev25["chunks"]),
            "INVALID_NB22D_POOL:" + q25,
        )

        gold25 = set(dev25["gold"][q25])
        require(bool(gold25), "EMPTY_GOLD:" + q25)

        replay25 = metrics(baseline25, gold25)
        require(
            all(
                abs(value - ranks25[q25]["metrics"][key]) <= 1e-12
                for key, value in replay25.items()
            ),
            "BASELINE_METRIC_REPLAY_CONFLICT:" + q25,
        )

        previous25 = prior_by_query25[q25]
        require(
            abs(
                replay25["recall_at_10"]
                - previous25["baseline_recall_at_10"]
            ) <= 1e-12,
            "C024_BASELINE_REPLAY_CONFLICT:" + q25,
        )

        head40 = set(baseline25[:40])
        gold40 = gold25 & head40
        ceiling40 = min(10, len(gold40)) / len(gold25)

        require(
            abs(
                ceiling40 - previous25["top40_recall_at_10_ceiling"]
            ) <= 1e-12,
            "C024_TOP40_CEILING_REPLAY_CONFLICT:" + q25,
        )

        for method25, candidates25 in (
            ("baseline_top40", baseline25[:40]),
            ("baseline_top100", baseline25[:100]),
            ("nb22d_full_pool", pool25),
        ):
            candidate_set25 = set(candidates25)
            available25 = gold25 & candidate_set25
            added25 = available25 - gold40
            lost25 = gold40 - available25

            ceiling25 = min(10, len(available25)) / len(gold25)
            outside25 = len(gold25 - candidate_set25) / len(gold25)
            capacity25 = max(0, len(available25) - 10) / len(gold25)

            require(
                abs(ceiling25 + outside25 + capacity25 - 1) <= 1e-12,
                "CEILING_DECOMPOSITION_CONFLICT:" + q25,
            )

            cases25.append({
                "query_id": q25,
                "cohort": previous25["cohort"],
                "source_types": previous25["source_types"],
                "candidate_method": method25,
                "candidate_count": len(candidates25),
                "contains_baseline_top40": head40.issubset(candidate_set25),
                "baseline_recall_at_10": replay25["recall_at_10"],
                "fusion_recall_at_10": previous25["fusion_recall_at_10"],
                "recall_at_10_ceiling": ceiling25,
                "ceiling_delta_vs_top40": ceiling25 - ceiling40,
                "outside_candidate_gap": outside25,
                "top10_capacity_gap": capacity25,
                "gold_count": len(gold25),
                "gold_in_candidates": len(available25),
                "outside_gold_count": len(gold25 - candidate_set25),
                "added_gold_count_vs_top40": len(added25),
                "lost_gold_count_vs_top40": len(lost25),
                "added_gold_fraction_vs_top40": len(added25) / len(gold25),
                "lost_gold_fraction_vs_top40": len(lost25) / len(gold25),
                "added_gold_ids_vs_top40": sorted(added25),
                "lost_gold_ids_vs_top40": sorted(lost25),
            })

    methods25 = (
        "baseline_top40", "baseline_top100", "nb22d_full_pool"
    )
    summaries25 = {}
    sources25 = {}

    for cohort_name25 in cohort_ids25:
        summaries25[cohort_name25] = {}
        sources25[cohort_name25] = {}

        for method25 in methods25:
            rows25 = [
                row for row in cases25
                if row["cohort"] == cohort_name25
                and row["candidate_method"] == method25
            ]
            require(len(rows25) == 60, "SUMMARY_QUERY_COUNT_CONFLICT")
            summaries25[cohort_name25][method25] = c025_mean(rows25)

        sources25[cohort_name25] = {}
        for source25 in ("book", "transcript", "website"):
            sources25[cohort_name25][source25] = {}
            for method25 in methods25:
                rows25 = [
                    row for row in cases25
                    if row["cohort"] == cohort_name25
                    and row["candidate_method"] == method25
                    and source25 in row["source_types"]
                ]
                require(len(rows25) == 20, "SOURCE_SLICE_COUNT_CONFLICT")
                sources25[cohort_name25][source25][method25] = c025_mean(rows25)

        require(
            abs(
                summaries25[cohort_name25]["baseline_top40"]
                ["recall_at_10_ceiling"]
                - coverage_report25["cohorts"][cohort_name25]
                ["top40_recall_at_10_ceiling"]
            ) <= 1e-12,
            "C024_AGGREGATE_CEILING_CONFLICT",
        )

    combined25 = {
        method25: c025_mean([
            row for row in cases25
            if row["candidate_method"] == method25
        ])
        for method25 in methods25
    }

    C025.update({
        "status": "COMPLETE",
        "query_count": 120,
        "baseline_binding": rank_binding25,
        "pool_binding": pool_binding25,
        "cohorts": summaries25,
        "per_source": sources25,
        "combined_descriptive_only": combined25,
        "interpretation_limit": (
            "Oracle Recall@10 capacity only. Full NB22D pool is not "
            "assumed to contain baseline Top100 or even baseline Top40. "
            "No model improvement or release acceptance is established."
        ),
        "next_step": "REVIEW_CAPACITY_GAIN_BEFORE_SELECTING_ONE_NEXT_EXPERIMENT",
        "scope": "DEVELOPMENT_ONLY_NOT_INDEPENDENT_ACCEPTANCE",
    })

    C025["cases_artifact"] = save(
        "C025.candidate_coverage_cases", cases25, rows=True
    )
    C025["report"] = save("C025.candidate_coverage_report", C025.copy())

    # Compact overall table.
    method_labels25 = {
        "baseline_top40": "Top40 baseline",
        "baseline_top100": "Top100 baseline",
        "nb22d_full_pool": "pool کامل NB22D",
    }
    main_rows25 = []

    for cohort_name25, cohort_label25 in (
        ("selection_dev60", "۶۰ سؤال اول"),
        ("additional_dev60", "۶۰ سؤال دوم"),
    ):
        for method25 in methods25:
            result25 = summaries25[cohort_name25][method25]
            tick25 = "✅" if result25["ceiling_has_target_capacity"] else ""
            main_rows25.append(
                f"<tr><td>{cohort_label25}</td>"
                f"<td>{method_labels25[method25]}</td>"
                f"<td>{result25['baseline_recall_at_10']:.2%}</td>"
                f"<td>{result25['fusion_recall_at_10']:.2%}</td>"
                f"<td>{result25['recall_at_10_ceiling']:.2%}</td>"
                f"<td>{100*result25['ceiling_delta_vs_top40']:+.2f}</td>"
                f"<td>{result25['queries_with_added_gold_vs_top40']}</td>"
                f"<td>{result25['queries_with_lost_gold_vs_top40']}</td>"
                f"<td>85%</td><td>{tick25}</td></tr>"
            )

    # Show source ceilings together for direct comparison.
    source_rows25 = []
    for cohort_name25, cohort_label25 in (
        ("selection_dev60", "۶۰ سؤال اول"),
        ("additional_dev60", "۶۰ سؤال دوم"),
    ):
        for source25, source_label25 in (
            ("book", "کتاب"),
            ("transcript", "متن صوتی"),
            ("website", "وب‌سایت"),
        ):
            group25 = sources25[cohort_name25][source25]
            source_rows25.append(
                f"<tr><td>{cohort_label25}</td><td>{source_label25}</td>"
                f"<td>{group25['baseline_top40']['recall_at_10_ceiling']:.2%}</td>"
                f"<td>{group25['baseline_top100']['recall_at_10_ceiling']:.2%}</td>"
                f"<td>{group25['nb22d_full_pool']['recall_at_10_ceiling']:.2%}</td>"
                "</tr>"
            )

    display(HTML(
        '<div dir="rtl" style="text-align:right">'
        "<p><strong>سقف نظری است، نه نتیجهٔ اجرای مدل.</strong> "
        "ستون «بعد» همان ترکیب ثابت Top40 قبلی است. "
        "تیک فقط ظرفیت نظری رسیدن به هدف کلی را نشان می‌دهد.</p>"
        '<table style="text-align:center">'
        "<tr><th>مجموعه</th><th>ورودی</th><th>قبل</th>"
        "<th>بعدِ فعلی</th><th>سقف Recall@10</th>"
        "<th>رشد سقف نسبت به Top40؛ واحد درصد</th>"
        "<th>سؤال با شاهد اضافه</th><th>سؤال با شاهد ازدست‌رفته</th>"
        "<th>هدف</th><th>ظرفیت کافی</th></tr>"
        + "".join(main_rows25)
        + "</table><br>"
        "<p>سقف‌ها به تفکیک منبع؛ هدف ۸۵٪ الزام جداگانهٔ هر منبع نیست.</p>"
        '<table style="text-align:center">'
        "<tr><th>مجموعه</th><th>منبع</th><th>Top40</th>"
        "<th>Top100</th><th>pool کامل NB22D</th></tr>"
        + "".join(source_rows25)
        + "</table></div>"
    ))

    print(json.dumps(C025, ensure_ascii=False, indent=2))

except Exception as exc:
    C025.update({"status": "FAILED", "error": str(exc)})
    print(json.dumps(C025, ensure_ascii=False, indent=2))
    raise
```

</ul>
</div>

```python
# C026 v1 — Preserve baseline Top40; add up to 60 pool candidates
# by round-robin over source_id groups.
# Selection is independent of Gold labels. Evaluation happens afterward.

import json
import math
from collections import OrderedDict
from IPython.display import display, HTML

C026 = {
    "version": "NB22E_DOCUMENT_DIVERSE_CANDIDATES_V1",
    "status": "PREPARING",
    "new_model_pairs": 0,
    "new_api_calls": 0,
    "downloads": False,
    "automatic_replacement": False,
    "release_acceptance_granted": False,
}


def c026_select(baseline_order, pool_order, chunk_map):
    """No Gold or evaluation outcomes enter this function."""
    head = list(baseline_order[:40])
    require(len(head) == len(set(head)) == 40, "INVALID_TOP40")

    head_set = set(head)
    groups = OrderedDict()

    for cid in pool_order:
        if cid in head_set:
            continue
        sid = chunk_map[cid].get("source_id")
        require(
            isinstance(sid, str) and sid.strip(),
            "SOURCE_ID_REQUIRED:" + cid,
        )
        groups.setdefault(sid, []).append(cid)

    # Source groups are ordered by their first appearance in the saved pool.
    additions = []
    level = 0
    while len(additions) < 60:
        progressed = False
        for ids in groups.values():
            if level < len(ids):
                additions.append(ids[level])
                progressed = True
                if len(additions) == 60:
                    break
        if not progressed:
            break
        level += 1

    return head + additions


def c026_summary(rows):
    require(bool(rows), "EMPTY_GROUP")
    keys = (
        "top40_ceiling", "top100_ceiling", "diverse_ceiling",
        "delta_vs_top100", "delta_vs_top40",
    )
    result = {
        key: math.fsum(row[key] for row in rows) / len(rows)
        for key in keys
    }
    result.update({
        "queries": len(rows),
        "candidate_count_min": min(r["candidate_count"] for r in rows),
        "candidate_count_max": max(r["candidate_count"] for r in rows),
        "queries_better_than_top100": sum(
            r["delta_vs_top100"] > 1e-12 for r in rows
        ),
        "queries_worse_than_top100": sum(
            r["delta_vs_top100"] < -1e-12 for r in rows
        ),
        "queries_unchanged_vs_top100": sum(
            abs(r["delta_vs_top100"]) <= 1e-12 for r in rows
        ),
        "oracle_recall_target_capacity": result["diverse_ceiling"] >= 0.85,
    })
    return result


try:
    require(C025.get("status") == "COMPLETE", "C025_REQUIRED")

    report26 = read_json(C025["report"])
    cases25_saved26 = read_rows(C025["cases_artifact"])

    reference26 = {}
    for row in cases25_saved26:
        qid = row["query_id"]
        method = row["candidate_method"]
        require(
            method not in reference26.setdefault(qid, {}),
            "DUPLICATE_C025_METHOD",
        )
        reference26[qid][method] = row

    needed26 = set(reference26)
    require(len(needed26) == 120, "EXPECTED_120_QUERIES")

    print("بررسی فایل‌های اصلی baseline و pool…", flush=True)
    ranks26 = c025_subset(report26["baseline_binding"], needed26)
    pools26 = c025_subset(report26["pool_binding"], needed26)

    specification26 = {
        "method": "BASELINE_TOP40_PLUS_POOL_DOCUMENT_ROUND_ROBIN60",
        "maximum_candidates": 100,
        "protected_baseline_prefix": 40,
        "maximum_additions": 60,
        "group_key": "source_id",
        "group_order": "FIRST_APPEARANCE_IN_SAVED_POOL",
        "within_group_order": "SAVED_POOL_ORDER",
        "selection_uses_gold": False,
        "selection_uses_query_source_type": False,
        "parameter_search": False,
    }
    C026["specification_artifact"] = save(
        "C026.candidate_selection_spec", specification26
    )

    candidates26 = {}
    evaluation_rows26 = []

    for qid in sorted(needed26):
        require(
            ranks26[qid]["split"] == pools26[qid]["split"] == "development",
            "SPLIT_CONFLICT",
        )
        baseline = ranks26[qid]["ranked_chunk_ids"]
        pool = pools26[qid]["candidate_chunk_ids"]

        require(
            len(baseline) >= 100
            and len(baseline) == len(set(baseline))
            and len(pool) == len(set(pool))
            and set(baseline).issubset(LAB["dev"]["chunks"])
            and set(pool).issubset(LAB["dev"]["chunks"]),
            "INVALID_CANDIDATE_INPUT:" + qid,
        )

        # Freeze this query's selection before consulting its Gold labels.
        selected = c026_select(baseline, pool, LAB["dev"]["chunks"])
        require(
            len(selected) == len(set(selected))
            and selected[:40] == baseline[:40]
            and 40 <= len(selected) <= 100,
            "SELECTION_CONTRACT_CONFLICT:" + qid,
        )
        candidates26[qid] = selected

        # Evaluation only; these labels do not feed back into selection.
        gold = set(LAB["dev"]["gold"][qid])
        require(bool(gold), "EMPTY_GOLD:" + qid)

        def ceiling(ids):
            return min(10, len(set(ids) & gold)) / len(gold)

        top40 = ceiling(baseline[:40])
        top100 = ceiling(baseline[:100])
        diverse = ceiling(selected)

        require(
            abs(
                top40
                - reference26[qid]["baseline_top40"]["recall_at_10_ceiling"]
            ) <= 1e-12
            and abs(
                top100
                - reference26[qid]["baseline_top100"]["recall_at_10_ceiling"]
            ) <= 1e-12,
            "C025_CEILING_REPLAY_CONFLICT:" + qid,
        )

        evaluation_rows26.append({
            "query_id": qid,
            "cohort": reference26[qid]["baseline_top40"]["cohort"],
            "source_types": reference26[qid]["baseline_top40"]["source_types"],
            "candidate_count": len(selected),
            "top40_ceiling": top40,
            "top100_ceiling": top100,
            "diverse_ceiling": diverse,
            "delta_vs_top100": diverse - top100,
            "delta_vs_top40": diverse - top40,
        })

    # Save selected candidates irrespective of whether their results improve.
    C026["candidate_artifact"] = save("C026.selected_candidates", {
        "specification": specification26,
        "candidates": candidates26,
    })

    cohorts26 = {}
    sources26 = {}

    for cohort in ("selection_dev60", "additional_dev60"):
        rows = [r for r in evaluation_rows26 if r["cohort"] == cohort]
        require(len(rows) == 60, "COHORT_COUNT_CONFLICT")
        cohorts26[cohort] = c026_summary(rows)
        sources26[cohort] = {}

        for source in ("book", "transcript", "website"):
            sliced = [r for r in rows if source in r["source_types"]]
            require(len(sliced) == 20, "SOURCE_COUNT_CONFLICT")
            sources26[cohort][source] = c026_summary(sliced)

    # This is only a capacity screen, not an acceptance gate.
    capacity_screen26 = all(
        result["delta_vs_top100"] >= -1e-12
        for result in cohorts26.values()
    ) and any(
        result["delta_vs_top100"] > 1e-12
        for result in cohorts26.values()
    )

    C026.update({
        "status": "COMPLETE",
        "specification": specification26,
        "cohorts": cohorts26,
        "per_source": sources26,
        "combined_descriptive_only": c026_summary(evaluation_rows26),
        "passes_exploratory_capacity_screen": capacity_screen26,
        "maximum_new_pairs_if_scored": sum(
            len(ids) - 40 for ids in candidates26.values()
        ),
        "inference_started": False,
        "scope": "ORACLE_CAPACITY_ONLY_DEVELOPMENT_NOT_ACCEPTANCE",
        "next_step": (
            "REVIEW_CAPACITY_AND_COST_BEFORE_ANY_MODEL_EXECUTION"
        ),
    })

    C026["cases_artifact"] = save(
        "C026.coverage_comparison_cases", evaluation_rows26, rows=True
    )
    C026["report"] = save("C026.coverage_comparison_report", C026.copy())

    table_rows = []
    for cohort, label in (
        ("selection_dev60", "۶۰ سؤال اول"),
        ("additional_dev60", "۶۰ سؤال دوم"),
    ):
        groups = [(label, cohorts26[cohort])]
        for source, source_label in (
            ("book", "کتاب"),
            ("transcript", "متن صوتی"),
            ("website", "وب‌سایت"),
        ):
            groups.append((label + "؛ " + source_label, sources26[cohort][source]))

        for group_label, result in groups:
            tick = "✅" if result["oracle_recall_target_capacity"] else ""
            table_rows.append(
                f"<tr><td>{group_label}</td>"
                f"<td>{result['top40_ceiling']:.2%}</td>"
                f"<td>{result['top100_ceiling']:.2%}</td>"
                f"<td>{result['diverse_ceiling']:.2%}</td>"
                f"<td>{100*result['delta_vs_top100']:+.2f}</td>"
                f"<td>{result['queries_better_than_top100']}</td>"
                f"<td>{result['queries_worse_than_top100']}</td>"
                f"<td>85%</td><td>{tick}</td></tr>"
            )

    display(HTML(
        '<div dir="rtl" style="text-align:right">'
        "<p><strong>تمام اعداد سقف نظری Recall@10 هستند؛ "
        "نتیجهٔ اجرای مدل نیستند.</strong> "
        "تیک فقط ظرفیت نظری هدف را نشان می‌دهد؛ "
        "هدف منابع جداگانه الزام پذیرش نیست.</p>"
        '<table style="text-align:center">'
        "<tr><th>مجموعه</th><th>Top40</th><th>Top100</th>"
        "<th>انتخاب متنوع</th>"
        "<th>تغییر نسبت به Top100؛ واحد درصد</th>"
        "<th>سؤال بهتر</th><th>سؤال بدتر</th>"
        "<th>هدف</th><th>ظرفیت کافی</th></tr>"
        + "".join(table_rows) + "</table></div>"
    ))

    print(json.dumps(C026, ensure_ascii=False, indent=2))

except Exception as exc:
    C026.update({"status": "FAILED", "error": str(exc)})
    print(json.dumps(C026, ensure_ascii=False, indent=2))
    raise
```

<h3>C027 — نسخهٔ ۱: آزمایش نهایی گسترش ورودی به Top100</h3>
<ul>
<li>همان ۱۲۰ سؤال؛ گزارش جداگانهٔ دو مجموعهٔ ۶۰تایی.</li>
<li>ورودی: ۱۰۰ نتیجهٔ اول baseline اصلی NB22D.</li>
<li>رتبه‌بندی BGE، سپس RRF با وزن برابر و k=60.</li>
<li>استفاده از کش Top40؛ حداکثر ۷۲۰۰ جفت جدید.</li>
<li>ذخیرهٔ هر جفت؛ امکان ادامه با اجرای مجدد همین سل.</li>
<li>بدون دانلود، API، تغییر وزن‌ها یا جایگزینی خودکار baseline.</li>
<li>نتیجهٔ مثبت نیز توسعه‌ای است؛ پذیرش مستقل ریلیز نیست.</li>

```python
# C027 v1 — Baseline Top100 -> BGE -> fixed equal-weight RRF(k=60).
# Reuses C017/C019/C021 pair cache. Does not mutate LAB.
# Soft time limit checked between pairs; cannot interrupt an active forward pass.

import json
import math
import time
import hashlib
from pathlib import Path
import torch
from IPython.display import display, HTML

C027 = {
    "version": "NB22E_BGE_TOP100_FIXED_RRF_V1",
    "status": "PREPARING",
    "downloads": False,
    "new_api_calls": 0,
    "new_model_pairs": 0,
    "automatic_replacement": False,
    "release_acceptance_granted": False,
}

try:
    require(C026.get("status") == "COMPLETE", "C026_REQUIRED")
    require(C023.get("status") == "COMPLETE", "C023_REQUIRED")

    coverage27 = read_json(C025["report"])
    reference27 = read_json(C023["report"])
    prior_orders27 = read_json(C023["orders_artifact"])
    frozen27 = read_json(C020["method_artifact"])

    score_spec27 = frozen27["score_contract"]
    score_hash27 = frozen27["score_contract_sha256"]
    require(
        c017_hash(score_spec27) == score_hash27,
        "SCORE_CONTRACT_CHECKSUM_CONFLICT",
    )

    runtime27 = BGE_RUNTIME
    require(
        runtime27["model_id"] == score_spec27["model_id"]
        and runtime27["revision"] == score_spec27["revision"]
        and runtime27["device"] == "cpu"
        and runtime27["dtype"] == "float32"
        and torch.__version__ == score_spec27["torch"]
        and score_spec27["batch_size"] == 1
        and score_spec27["max_length"] == 512,
        "RUNTIME_CONTRACT_CONFLICT",
    )

    prior_by_query27 = {}
    for row in prior_orders27["query_metrics"]:
        qid = row["query_id"]
        require(qid not in prior_by_query27, "DUPLICATE_QUERY")
        prior_by_query27[qid] = row

    cohorts27 = {
        name: sorted(
            q for q, row in prior_by_query27.items()
            if row["cohort"] == name
        )
        for name in ("selection_dev60", "additional_dev60")
    }
    require(
        len(prior_by_query27) == 120
        and all(len(ids) == 60 for ids in cohorts27.values()),
        "COHORT_SIZE_CONFLICT",
    )
    ids27 = cohorts27["selection_dev60"] + cohorts27["additional_dev60"]

    print("بررسی baseline و کش ۴۸۰۰ جفت قبلی…", flush=True)
    rank_rows27 = c025_subset(
        coverage27["baseline_binding"], set(ids27)
    )

    dev27 = LAB["dev"]
    questions27 = {
        q: golden[q]["evaluation_content"]["question"] for q in ids27
    }
    baseline27 = {}
    before27 = {}

    cache_dir27 = path_for(
        "artifacts/phase2/quality_improvement/NB22E.bge_top20_cache_v1/"
        + score_hash27
    )
    require(cache_dir27.is_dir(), "EXISTING_BGE_CACHE_REQUIRED")

    def c027_identity(qid, cid):
        return {
            "contract_sha256": score_hash27,
            "query_id": qid,
            "chunk_id": cid,
            "question_sha256": hashlib.sha256(
                questions27[qid].encode("utf-8")
            ).hexdigest(),
            "passage_sha256": hashlib.sha256(
                dev27["chunks"][cid]["content"].encode("utf-8")
            ).hexdigest(),
        }

    def c027_read_score(identity):
        path = cache_dir27 / (c017_hash(identity) + ".json")
        if not path.is_file():
            return path, None

        record = decode(path.read_bytes())
        body = record["body"]
        require(
            body["identity"] == identity
            and record["body_sha256"] == c017_hash(body),
            "PAIR_CACHE_CHECKSUM_OR_IDENTITY_CONFLICT",
        )
        score = float(body["score"])
        require(math.isfinite(score), "NONFINITE_CACHED_SCORE")
        return path, score

    # Complete preflight before any new inference.
    for qid in ids27:
        row = rank_rows27[qid]
        order = row["ranked_chunk_ids"]
        require(
            row["split"] == "development"
            and len(order) >= 100
            and len(order) == len(set(order))
            and set(order).issubset(dev27["chunks"]),
            "BASELINE_ORDER_CONFLICT:" + qid,
        )
        require(
            isinstance(questions27[qid], str)
            and questions27[qid].strip(),
            "INVALID_QUESTION:" + qid,
        )

        measured = metrics(order, dev27["gold"][qid])
        require(
            all(
                abs(v - row["metrics"][k]) <= 1e-12
                for k, v in measured.items()
            ),
            "BASELINE_REPLAY_CONFLICT:" + qid,
        )

        baseline27[qid] = list(order)
        before27[qid] = {
            k: measured[k]
            for k in ("recall_at_10", "ndcg_at_10", "hit_at_10")
        }

        for key, value in before27[qid].items():
            require(
                abs(value - prior_by_query27[qid]["before"][key]) <= 1e-9,
                "PRIOR_BASELINE_METRIC_CONFLICT:" + qid,
            )

        for cid in order[:100]:
            passage = dev27["chunks"][cid]["content"]
            require(
                isinstance(passage, str) and passage.strip(),
                "INVALID_PASSAGE:" + cid,
            )

        for cid in order[:40]:
            _, score = c027_read_score(c027_identity(qid, cid))
            require(
                score is not None,
                "TOP40_CACHE_MISSING_NO_NEW_INFERENCE_STARTED:" + qid,
            )

    specification27 = {
        "method": "BASELINE_TOP100_BGE_EQUAL_RRF",
        "depth": 100,
        "rrf_k": 60,
        "baseline_weight": 0.5,
        "bge_weight": 0.5,
        "tie_break": "BASELINE_POSITION",
        "tail": "UNCHANGED_BASELINE_AFTER_TOP100",
        "score_contract_sha256": score_hash27,
        "maximum_new_pairs": 7200,
        "soft_time_limit_seconds": 21600,
        "parameter_search": False,
    }
    C027["plan_artifact"] = save("C027.top100_fixed_plan", {
        "specification": specification27,
        "cohorts": cohorts27,
        "baseline_binding": coverage27["baseline_binding"],
        "prior_top40_report": C023["report"],
        "score_contract": score_spec27,
    })

    print(
        "PASS: baseline و کش Top40 تأیید شدند؛ شروع امتیازدهی Top100.",
        flush=True,
    )

    model27 = runtime27["model"]
    tokenizer27 = runtime27["tokenizer"]
    model27.eval()
    torch.set_num_threads(score_spec27["threads"])

    started27 = time.perf_counter()
    new27 = 0
    cached27 = 0
    orders27 = {}
    after27 = {}
    pure_bge27 = {}

    for index, qid in enumerate(ids27, start=1):
        head = baseline27[qid][:100]
        positions = {cid: i + 1 for i, cid in enumerate(head)}
        scores = {}

        for cid in head:
            require(
                time.perf_counter() - started27 < 21600,
                "SOFT_TIME_LIMIT_REACHED_CACHE_PRESERVED_RERUN_THIS_CELL",
            )
            identity = c027_identity(qid, cid)
            path, score = c027_read_score(identity)

            if score is None:
                require(new27 < 7200, "NEW_PAIR_LIMIT_REACHED")
                encoded = tokenizer27(
                    questions27[qid],
                    dev27["chunks"][cid]["content"],
                    return_tensors="pt",
                    truncation=True,
                    max_length=512,
                    padding=False,
                )
                with torch.inference_mode():
                    logits = model27(**encoded).logits

                require(logits.numel() == 1, "INVALID_OUTPUT_SHAPE")
                score = float(logits.item())
                require(math.isfinite(score), "NONFINITE_MODEL_SCORE")

                body = {
                    "identity": identity,
                    "score": score,
                    "input_tokens": int(encoded["input_ids"].shape[1]),
                }
                c017_atomic_json(path, {
                    "body": body,
                    "body_sha256": c017_hash(body),
                })
                new27 += 1
                C027["new_model_pairs"] = new27
            else:
                cached27 += 1

            scores[cid] = score

        bge_head = sorted(
            head, key=lambda cid: (-scores[cid], positions[cid])
        )
        bge_ranks = {cid: i + 1 for i, cid in enumerate(bge_head)}
        fused_head = sorted(
            head,
            key=lambda cid: (
                -(0.5 / (60 + positions[cid])
                  + 0.5 / (60 + bge_ranks[cid])),
                positions[cid],
            ),
        )

        orders27[qid] = fused_head + baseline27[qid][100:]
        after27[qid] = c017_metrics(
            orders27[qid], dev27["gold"][qid]
        )
        pure_bge27[qid] = c017_metrics(
            bge_head + baseline27[qid][100:], dev27["gold"][qid]
        )

        elapsed = time.perf_counter() - started27
        print(json.dumps({
            "queries_completed": index,
            "queries_total": 120,
            "cohort": prior_by_query27[qid]["cohort"],
            "new_pairs": new27,
            "cached_pairs": cached27,
            "elapsed_minutes": round(elapsed / 60, 2),
            "rough_remaining_minutes": round(
                elapsed / index * (120 - index) / 60, 2
            ),
        }), flush=True)

    results27 = {}
    for name, ids in cohorts27.items():
        result = c022_summarize(ids, before27, after27)
        result["top40_fixed_fusion"] = reference27["cohorts"][name]["after"]
        result["pure_bge_top100"] = c022_mean(pure_bge27[q] for q in ids)
        result["passes_point_guard"] = (
            result["aggregate_no_regression"]
            and result["after"]["ndcg_at_10"]
            > result["before"]["ndcg_at_10"] + 1e-12
        )
        result["per_source"] = {}
        for source in ("book", "transcript", "website"):
            source_ids = [
                q for q in ids if source in golden[q]["source_types"]
            ]
            result["per_source"][source] = c022_summarize(
                source_ids, before27, after27
            )
        results27[name] = result

    C027.update({
        "status": "COMPLETE",
        "specification": specification27,
        "cohorts": results27,
        "combined_descriptive_only": c022_summarize(
            ids27, before27, after27
        ),
        "passes_point_guard_in_both_cohorts": all(
            r["passes_point_guard"] for r in results27.values()
        ),
        "absolute_targets_met_in_both_cohorts": all(
            r["absolute_quality_targets_met"] for r in results27.values()
        ),
        "new_model_pairs": new27,
        "cached_pairs": cached27,
        "wall_seconds": time.perf_counter() - started27,
        "scope": "POST_OUTCOME_DEVELOPMENT_EXPERIMENT_NOT_INDEPENDENT_ACCEPTANCE",
    })

    C027["orders_artifact"] = save("C027.top100_fixed_orders", {
        "specification": specification27,
        "cohorts": cohorts27,
        "orders": orders27,
        "before_metrics": before27,
        "after_metrics": after27,
        "pure_bge_metrics": pure_bge27,
    })
    C027["report"] = save("C027.top100_fixed_report", C027.copy())

    html_rows = []
    for name, label in (
        ("selection_dev60", "۶۰ سؤال اول"),
        ("additional_dev60", "۶۰ سؤال دوم"),
    ):
        result = results27[name]
        for key, metric_label, target in (
            ("recall_at_10", "Recall@10", 0.85),
            ("ndcg_at_10", "nDCG@10", 0.75),
            ("hit_at_10", "Hit@10", None),
        ):
            threshold = "تعیین نشده" if target is None else f"{target:.0%}"
            tick = (
                "✅"
                if target is not None and result["after"][key] >= target
                else ""
            )
            html_rows.append(
                f"<tr><td>{label}</td><td>{metric_label}</td>"
                f"<td>{result['before'][key]:.2%}</td>"
                f"<td>{result['top40_fixed_fusion'][key]:.2%}</td>"
                f"<td>{result['after'][key]:.2%}</td>"
                f"<td>{result['delta_percentage_points'][key]:+.2f}</td>"
                f"<td>{threshold}</td><td>{tick}</td></tr>"
            )

    display(HTML(
        '<div dir="rtl" style="text-align:right">'
        "<p>نتایج توسعه‌ای؛ پذیرش مستقل نیست. "
        "تیک فقط رسیدن به حد عددی همان معیار است.</p>"
        '<table style="text-align:center">'
        "<tr><th>مجموعه</th><th>معیار</th><th>قبل: baseline</th>"
        "<th>ترکیب Top40</th><th>بعد: ترکیب Top100</th>"
        "<th>تغییر نسبت به baseline؛ واحد درصد</th>"
        "<th>حد پذیرش</th><th>قبول</th></tr>"
        + "".join(html_rows) + "</table></div>"
    ))
    print(json.dumps(C027, ensure_ascii=False, indent=2))

except KeyboardInterrupt:
    C027.update({
        "status": "INTERRUPTED_RESUMABLE",
        "message": "کش حفظ شده؛ برای ادامه همین سل را دوباره اجرا کن.",
    })
    print(json.dumps(C027, ensure_ascii=False, indent=2))
    raise

except Exception as exc:
    C027.update({
        "status": "FAILED_OR_LIMIT_REACHED_CACHE_PRESERVED",
        "error": str(exc),
    })
    print(json.dumps(C027, ensure_ascii=False, indent=2))
    raise
```

<h3>C028 — نسخهٔ ۱: جمع‌بندی نهایی آزمایش‌های BGE</h3>
<ul>
<li>ثبت مقایسهٔ baseline، BGE خالص، ترکیب Top40 و ترکیب Top100.</li>
<li>انتخاب ترکیب ثابت Top40 فقط به‌عنوان گزینهٔ توسعه‌ای.</li>
<li>حفظ baseline؛ هدف‌های کیفیت هنوز کامل تأمین نشده‌اند.</li>
<li>بستن شاخهٔ افزایش عمق و تنظیم وزن در این دور آزمایش.</li>
<li>حفظ تمام کش‌ها و ثبت مسیر گزارش‌های مرجع.</li>
<li>گام بعدی: طراحی تغییر در بازیابی یا انتخاب نامزد، با بودجهٔ محدود.</li>
<li>هیچ اجرای مدل، دانلود، API یا پذیرش ریلیز انجام نمی‌شود.</li>
</ul>
</div>

```python
# C028 v1 — Final scoped reporting; no inference or model loading.

import json
import math
from IPython.display import display, HTML

C028 = {
    "version": "NB22E_BGE_EXPERIMENT_CLOSEOUT_V1",
    "status": "PREPARING",
    "new_model_pairs": 0,
    "new_api_calls": 0,
    "downloads": False,
    "automatic_replacement": False,
    "release_acceptance_granted": False,
}

try:
    report_variables28 = {
        "pure_bge_first60": C019,
        "pure_bge_next60": C021,
        "two_cohort_audit": C022,
        "fixed_top40": C023,
        "top40_gap": C024,
        "candidate_coverage": C025,
        "document_diversity": C026,
        "fixed_top100": C027,
    }

    bindings28 = {}
    reports28 = {}

    print("بررسی هش گزارش‌های نهایی؛ بدون تکرار محاسبات مدل…", flush=True)

    for name28, variable28 in report_variables28.items():
        require(
            variable28.get("status") == "COMPLETE",
            "COMPLETE_REPORT_REQUIRED:" + name28,
        )
        bindings28[name28] = variable28["report"]
        reports28[name28] = read_json(variable28["report"])
        require(
            reports28[name28]["status"] == "COMPLETE",
            "SAVED_REPORT_NOT_COMPLETE:" + name28,
        )

    fixed40 = reports28["fixed_top40"]
    fixed100 = reports28["fixed_top100"]
    audit28 = reports28["two_cohort_audit"]
    coverage28 = reports28["candidate_coverage"]

    require(
        audit28["all_saved_metrics_reproduced"]
        and audit28["all_top40_order_contracts_verified"],
        "VERIFIED_AUDIT_REQUIRED",
    )

    method_specs28 = (
        fixed40["specification"],
        fixed100["specification"],
    )
    require(
        all(
            spec["rrf_k"] == 60
            and spec["baseline_weight"] == 0.5
            and spec["bge_weight"] == 0.5
            for spec in method_specs28
        ),
        "FIXED_FUSION_PARAMETERS_CONFLICT",
    )
    require(
        fixed40["specification"]["depth"] == 40
        and fixed100["specification"]["depth"] == 100,
        "DEPTH_CONTRACT_CONFLICT",
    )

    comparisons28 = {}
    pooled_values28 = {
        method: {"before": [], "after": []}
        for method in ("fixed_top40", "fixed_top100")
    }

    for cohort28, pure_report_name28 in (
        ("selection_dev60", "pure_bge_first60"),
        ("additional_dev60", "pure_bge_next60"),
    ):
        top40 = fixed40["cohorts"][cohort28]
        top100 = fixed100["cohorts"][cohort28]
        pure = reports28[pure_report_name28]

        require(
            top40["queries"] == top100["queries"] == 60,
            "COHORT_COUNT_CONFLICT",
        )

        for key28 in ("recall_at_10", "ndcg_at_10", "hit_at_10"):
            require(
                abs(top40["before"][key28] - top100["before"][key28])
                <= 1e-9
                and abs(top40["before"][key28] - pure["before"][key28])
                <= 1e-9,
                "BASELINE_METRIC_CONFLICT:" + cohort28 + "/" + key28,
            )

        comparisons28[cohort28] = {
            "queries": 60,
            "baseline": top40["before"],
            "pure_bge_top40": pure["after"],
            "fixed_top40": top40["after"],
            "fixed_top100": top100["after"],
            "top100_minus_top40_percentage_points": {
                key: 100 * (top100["after"][key] - top40["after"][key])
                for key in top40["after"]
            },
        }

        for method28, result28 in (
            ("fixed_top40", top40),
            ("fixed_top100", top100),
        ):
            pooled_values28[method28]["before"].append(result28["before"])
            pooled_values28[method28]["after"].append(result28["after"])

    # Equal-sized cohorts: the arithmetic mean equals the pooled macro mean.
    combined28 = {}
    for method28, values28 in pooled_values28.items():
        combined28[method28] = {}
        for stage28 in ("before", "after"):
            combined28[method28][stage28] = {
                key: math.fsum(row[key] for row in values28[stage28]) / 2
                for key in ("recall_at_10", "ndcg_at_10", "hit_at_10")
            }

    C028.update({
        "status": "COMPLETE",
        "scope": "BGE_EXPERIMENT_BRANCH_CLOSEOUT_NOT_NB22_OR_PHASE2_COMPLETION",
        "reference_reports": bindings28,
        "comparisons": comparisons28,
        "combined_descriptive_only": combined28,
        "development_candidate": {
            "method": "EQUAL_WEIGHT_RRF_BASELINE_BGE_TOP40",
            "rrf_k": 60,
            "baseline_weight": 0.5,
            "bge_weight": 0.5,
            "depth": 40,
            "specification_artifact": fixed40["specification_artifact"],
            "report": bindings28["fixed_top40"],
            "reason": (
                "Improved aggregate nDCG with no aggregate Recall/Hit loss "
                "in both development cohorts; Top100 added little benefit."
            ),
            "accepted_for_release": False,
        },
        "current_baseline_action": "KEEP_EXISTING_BASELINE",
        "quality_targets": {
            "recall_at_10": 0.85,
            "ndcg_at_10": 0.75,
            "hit_at_10": None,
        },
        "quality_acceptance": "NOT_GRANTED",
        "closed_experiments": [
            "PURE_BGE_TOP40_AS_BASELINE_REPLACEMENT_NOT_SELECTED",
            "DOCUMENT_ROUND_ROBIN_SELECTION_NOT_SELECTED",
            "TOP100_EXPANSION_NOT_SELECTED_OVER_FIXED_TOP40",
            "NO_MORE_DEPTH_OR_WEIGHT_SEARCH_IN_THIS_EXPERIMENT_ROUND",
        ],
        "preserve": [
            "EXISTING_BASELINE",
            "ALL_BGE_PAIR_SCORE_CACHES",
            "PINNED_LOCAL_MODEL_FILES",
            "HASHED_REPORTS_AND_ORDER_ARTIFACTS",
        ],
        "known_limits": [
            "BOTH_COHORTS_ARE_DEVELOPMENT_DATA",
            "METHOD_SELECTION_USED_OBSERVED_DEVELOPMENT_OUTCOMES",
            "SOURCE_COMPONENT_INDEPENDENCE_NOT_ESTABLISHED",
            "ABSOLUTE_QUALITY_TARGETS_NOT_FULLY_MET",
            "AGGREGATE_NO_REGRESSION_DOES_NOT_MEAN_NO_QUERY_REGRESSIONS",
        ],
        "next_work": {
            "task": "DESIGN_ONE_MATERIALLY_DIFFERENT_RETRIEVAL_OR_SELECTION_METHOD",
            "input": bindings28["candidate_coverage"],
            "full_pool_recall_ceiling_descriptive": (
                coverage28["combined_descriptive_only"]
                ["nb22d_full_pool"]["recall_at_10_ceiling"]
            ),
            "requirements": [
                "USE_ACTUAL_SAVED_RETRIEVAL_COMPONENTS",
                "DO_NOT_USE_GOLD_LABELS_FOR_CANDIDATE_SELECTION",
                "DECLARE_COMPUTE_COST_BEFORE_EXECUTION",
                "REQUIRE_A_SMALL_PILOT_BEFORE_A_LARGE_RUN",
                "NO_NEW_MODEL_DOWNLOAD_BY_DEFAULT",
            ],
            "execution_started": False,
        },
        "future_runtime_preference": {
            "mode": "LOW_IMPACT",
            "initial_cpu_threads": 2,
            "bounded_sessions_minutes": [10, 15],
            "checkpoint_and_resume": True,
            "brief_pauses_between_inference_steps": True,
            "score_contract_change_requires_separate_cache_or_verification": True,
            "announce_estimated_duration_if_over_minutes": 5,
        },
    })

    C028["report"] = save("C028.bge_branch_closeout_report", C028.copy())

    table_rows28 = []
    for cohort28, label28 in (
        ("selection_dev60", "۶۰ سؤال اول"),
        ("additional_dev60", "۶۰ سؤال دوم"),
    ):
        result28 = comparisons28[cohort28]
        for key28, metric_label28, target28 in (
            ("recall_at_10", "Recall@10", 0.85),
            ("ndcg_at_10", "nDCG@10", 0.75),
            ("hit_at_10", "Hit@10", None),
        ):
            before28 = result28["baseline"][key28]
            after28 = result28["fixed_top40"][key28]
            tick28 = (
                "✅" if target28 is not None and after28 >= target28 else ""
            )
            threshold28 = (
                "تعیین نشده" if target28 is None else f"{target28:.0%}"
            )
            delta28 = 100 * (after28 - before28)
            if abs(delta28) < 1e-10:
                delta28 = 0.0

            table_rows28.append(
                f"<tr><td>{label28}</td><td>{metric_label28}</td>"
                f"<td>{before28:.2%}</td><td>{after28:.2%}</td>"
                f"<td>{delta28:+.2f}</td>"
                f"<td>{threshold28}</td><td>{tick28}</td></tr>"
            )

    display(HTML(
        '<div dir="rtl" style="text-align:right">'
        "<p><strong>شاخهٔ آزمایش BGE جمع‌بندی شد؛ "
        "NB22 و پذیرش کیفیت هنوز بسته نشده‌اند.</strong></p>"
        "<p>گزینهٔ توسعه‌ای: ترکیب ثابت Top40. "
        "baseline حفظ می‌شود؛ اجرای سنگین جدید آغاز نشده است.</p>"
        '<table style="text-align:center">'
        "<tr><th>مجموعه</th><th>معیار</th><th>قبل</th>"
        "<th>بعدِ ترکیب Top40</th><th>تغییر؛ واحد درصد</th>"
        "<th>حد پذیرش</th><th>قبول عددی</th></tr>"
        + "".join(table_rows28)
        + "</table></div>"
    ))

    print(json.dumps(C028, ensure_ascii=False, indent=2))

except Exception as exc:
    C028.update({"status": "FAILED", "error": str(exc)})
    print(json.dumps(C028, ensure_ascii=False, indent=2))
    raise
```

<h3>C029 — نسخهٔ ۱: انتخاب متوازن از مسیرهای واقعی بازیابی</h3>

<li>حفظ ۴۰ نتیجهٔ اول baseline.</li>

<li>انتخاب بدون استفاده از Golden؛ شاهدها فقط برای ارزیابی پوشش استفاده می‌شوند.</li>
<li>مقایسه با Top100 در بودجهٔ حداکثر ۱۰۰ نامزد.</li>

<li>سقف پوشش نظری است؛ پذیرش کیفیت یا تضمین عملکرد BGE نیست.</li>

```python
# C029 v1 — Protected Top40 + alternating word/char candidates.
# Reads the actual NB22D lexical-channel artifact.
# No execution of NB22D cells and no modification of CONFIG or LAB.

import json
import math
from IPython.display import display, HTML

C029 = {
    "version": "NB22E_BALANCED_LEXICAL_CANDIDATES_V1",
    "status": "PREPARING",
    "new_model_pairs": 0,
    "new_api_calls": 0,
    "downloads": False,
    "automatic_replacement": False,
    "release_acceptance_granted": False,
}


def c029_select(baseline, channels):
    # Gold labels are deliberately not an argument.
    selected = list(baseline[:40])
    seen = set(selected)
    positions = {"word": 0, "char": 0}
    additions = {"word": 0, "char": 0}

    while len(selected) < 100:
        progressed = False

        for channel in ("word", "char"):
            order = channels[channel]
            position = positions[channel]

            while position < len(order) and order[position] in seen:
                position += 1

            if position < len(order):
                cid = order[position]
                selected.append(cid)
                seen.add(cid)
                additions[channel] += 1
                position += 1
                progressed = True

            positions[channel] = position
            if len(selected) == 100:
                break

        if not progressed:
            break

    return selected, additions


def c029_summary(rows):
    require(bool(rows), "EMPTY_SUMMARY")
    result = {
        key: math.fsum(row[key] for row in rows) / len(rows)
        for key in (
            "top40_ceiling", "top100_ceiling",
            "balanced_ceiling", "delta_vs_top100",
        )
    }
    result.update({
        "queries": len(rows),
        "candidate_count_min": min(r["candidate_count"] for r in rows),
        "candidate_count_max": max(r["candidate_count"] for r in rows),
        "queries_better_than_top100": sum(
            r["delta_vs_top100"] > 1e-12 for r in rows
        ),
        "queries_worse_than_top100": sum(
            r["delta_vs_top100"] < -1e-12 for r in rows
        ),
        "queries_unchanged_vs_top100": sum(
            abs(r["delta_vs_top100"]) <= 1e-12 for r in rows
        ),
        "oracle_target_capacity": result["balanced_ceiling"] >= 0.85,
    })
    return result


try:
    require(C028.get("status") == "COMPLETE", "C028_REQUIRED")

    coverage29 = read_json(C025["report"])
    coverage_cases29 = read_rows(C025["cases_artifact"])
    d_final29 = read_json(PINS["D_final"])

    # Locate the actual hashed lexical-channel artifact, not a guessed filename.
    channel_bindings29 = {}
    for binding in d_final29["artifacts"]:
        if binding["path"].split("/")[-1].startswith("A03.lexical_channels."):
            channel_bindings29[
                (binding["path"], binding_hash(binding))
            ] = binding

    require(
        len(channel_bindings29) == 1,
        "UNAMBIGUOUS_NB22D_LEXICAL_CHANNEL_ARTIFACT_REQUIRED",
    )
    channel_binding29 = next(iter(channel_bindings29.values()))

    reference29 = {}
    for row in coverage_cases29:
        if row["candidate_method"] == "baseline_top100":
            qid = row["query_id"]
            require(qid not in reference29, "DUPLICATE_REFERENCE_QUERY")
            reference29[qid] = row

    needed29 = set(reference29)
    require(len(needed29) == 120, "EXPECTED_120_QUERIES")

    print("بررسی هش baseline و فهرست‌های واقعی word/char…", flush=True)
    ranks29 = c025_subset(coverage29["baseline_binding"], needed29)
    channels29 = c025_subset(channel_binding29, needed29)

    specification29 = {
        "method": "BASELINE_TOP40_PLUS_ALTERNATING_WORD_CHAR60",
        "maximum_candidates": 100,
        "protected_baseline_prefix": 40,
        "channel_cycle": ["word", "char"],
        "within_channel_order": "ORIGINAL_NB22D_ORDER",
        "duplicate_rule": "SKIP_ALREADY_SELECTED",
        "exhausted_channel_rule": "CONTINUE_WITH_OTHER_CHANNEL",
        "selection_uses_gold": False,
        "selection_uses_query_source_type": False,
        "parameter_search": False,
        "channel_binding": channel_binding29,
    }
    C029["specification_artifact"] = save(
        "C029.balanced_channels_spec", specification29
    )

    selected29 = {}
    cases29 = []

    for qid in sorted(needed29):
        require(ranks29[qid]["split"] == "development", "SPLIT_CONFLICT")
        baseline = ranks29[qid]["ranked_chunk_ids"]
        channels = channels29[qid]["channels"]

        require(
            len(baseline) >= 100
            and len(baseline) == len(set(baseline))
            and set(baseline).issubset(LAB["dev"]["chunks"]),
            "INVALID_BASELINE:" + qid,
        )
        require(set(channels) == {"word", "char"}, "CHANNEL_NAMES_CONFLICT")

        for order in channels.values():
            require(
                len(order) <= 800
                and len(order) == len(set(order))
                and set(order).issubset(LAB["dev"]["chunks"]),
                "INVALID_LEXICAL_CHANNEL:" + qid,
            )

        candidates, additions = c029_select(baseline, channels)
        require(
            candidates[:40] == baseline[:40]
            and len(candidates) == len(set(candidates))
            and 40 <= len(candidates) <= 100,
            "SELECTION_CONTRACT_CONFLICT",
        )
        selected29[qid] = candidates

        # Evaluation starts only after candidate selection is complete.
        gold = set(LAB["dev"]["gold"][qid])
        require(bool(gold), "EMPTY_GOLD:" + qid)

        def ceiling(order):
            return min(10, len(set(order) & gold)) / len(gold)

        top40 = ceiling(baseline[:40])
        top100 = ceiling(baseline[:100])
        balanced = ceiling(candidates)

        require(
            abs(
                top100 - reference29[qid]["recall_at_10_ceiling"]
            ) <= 1e-12,
            "TOP100_CEILING_REPLAY_CONFLICT",
        )

        cases29.append({
            "query_id": qid,
            "cohort": reference29[qid]["cohort"],
            "source_types": reference29[qid]["source_types"],
            "candidate_count": len(candidates),
            "additions_by_channel": additions,
            "top40_ceiling": top40,
            "top100_ceiling": top100,
            "balanced_ceiling": balanced,
            "delta_vs_top100": balanced - top100,
        })

    C029["candidate_artifact"] = save("C029.balanced_candidates", {
        "specification": specification29,
        "candidates": selected29,
    })

    cohorts29 = {}
    slices29 = {}

    for name in ("selection_dev60", "additional_dev60"):
        rows = [r for r in cases29 if r["cohort"] == name]
        require(len(rows) == 60, "COHORT_COUNT_CONFLICT")
        cohorts29[name] = c029_summary(rows)
        slices29[name] = {}

        for source in ("book", "transcript", "website"):
            source_rows = [r for r in rows if source in r["source_types"]]
            require(len(source_rows) == 20, "SOURCE_COUNT_CONFLICT")
            slices29[name][source] = c029_summary(source_rows)

    capacity_screen29 = (
        all(r["delta_vs_top100"] >= -1e-12 for r in cohorts29.values())
        and any(r["delta_vs_top100"] > 1e-12 for r in cohorts29.values())
    )

    C029.update({
        "status": "COMPLETE",
        "cohorts": cohorts29,
        "per_source": slices29,
        "combined_descriptive_only": c029_summary(cases29),
        "passes_exploratory_capacity_screen": capacity_screen29,
        "additional_candidate_slots_over_top40": sum(
            len(order) - 40 for order in selected29.values()
        ),
        "new_model_pairs_needed_not_yet_measured": True,
        "inference_started": False,
        "scope": "DEVELOPMENT_ORACLE_CAPACITY_ONLY_NOT_ACCEPTANCE",
    })
    C029["cases_artifact"] = save(
        "C029.balanced_coverage_cases", cases29, rows=True
    )
    C029["report"] = save("C029.balanced_coverage_report", C029.copy())

    html_rows29 = []
    for name, label in (
        ("selection_dev60", "۶۰ سؤال اول"),
        ("additional_dev60", "۶۰ سؤال دوم"),
    ):
        groups = [(label, cohorts29[name])]
        for source, source_label in (
            ("book", "کتاب"),
            ("transcript", "متن صوتی"),
            ("website", "وب‌سایت"),
        ):
            groups.append((label + "؛ " + source_label, slices29[name][source]))

        for group_label, result in groups:
            tick = "✅" if result["oracle_target_capacity"] else ""
            html_rows29.append(
                f"<tr><td>{group_label}</td>"
                f"<td>{result['top40_ceiling']:.2%}</td>"
                f"<td>{result['top100_ceiling']:.2%}</td>"
                f"<td>{result['balanced_ceiling']:.2%}</td>"
                f"<td>{100*result['delta_vs_top100']:+.2f}</td>"
                f"<td>{result['queries_better_than_top100']}</td>"
                f"<td>{result['queries_worse_than_top100']}</td>"
                f"<td>85%</td><td>{tick}</td></tr>"
            )

    display(HTML(
        '<div dir="rtl" style="text-align:right">'
        "<p><strong>اعداد، سقف نظری Recall@10 هستند؛ "
        "اجرای مدل انجام نشده است.</strong> "
        "تیک ظرفیت نظری را نشان می‌دهد، نه پذیرش واقعی. "
        "هدف ۸۵٪ الزام جداگانهٔ هر منبع نیست.</p>"
        '<table style="text-align:center">'
        "<tr><th>مجموعه</th><th>Top40</th><th>Top100</th>"
        "<th>انتخاب word/char</th><th>تغییر نسبت به Top100؛ واحد درصد</th>"
        "<th>سؤال بهتر</th><th>سؤال بدتر</th>"
        "<th>هدف</th><th>ظرفیت کافی</th></tr>"
        + "".join(html_rows29) + "</table></div>"
    ))

    print(json.dumps(C029, ensure_ascii=False, indent=2))

except Exception as exc:
    C029.update({"status": "FAILED", "error": str(exc)})
    print(json.dumps(C029, ensure_ascii=False, indent=2))
    raise
```

<h3>C030 — نسخهٔ ۱: شناسایی پیاده‌سازی و فایل‌های بازیابی معنایی</h3>
<ul>
<li>بررسی کد نوت‌بوک‌ها و فایل‌های Python مرتبط با embedding و جست‌وجوی برداری.</li>
<li>بررسی نام و اندازهٔ فایل‌های موجود در پوشه‌های embedding و ایندکس.</li>
<li>بدون خواندن خروجی سل‌ها یا بارگذاری فایل‌های مدل و بردار.</li>
<li>وجود فایل به معنی تأیید سلامت، سازگاری یا آمادگی اجرا نیست.</li>
<li>هیچ ساخت embedding، دانلود یا جست‌وجوی معنایی اجرا نمی‌شود.</li>
</ul>

```python
# C030 v1 — Bounded local inventory of semantic-retrieval code and files.
# No model, vector, or index loading. No network calls.

import json
import re
import time
from pathlib import Path

C030 = {
    "version": "NB22E_SEMANTIC_RETRIEVAL_INVENTORY_V1",
    "status": "PREPARING",
    "new_model_pairs": 0,
    "new_api_calls": 0,
    "downloads": False,
    "automatic_replacement": False,
    "release_acceptance_granted": False,
}

try:
    root30 = Path(ROOT).resolve()
    started30 = time.monotonic()

    # Metadata scan is bounded; skipped files are reported.
    max_code_bytes30 = 10 * 1024 * 1024
    max_seconds30 = 120
    max_files30 = 10000
    scanned30 = 0
    issues30 = []
    code_matches30 = []
    assets30 = []
    stopped30 = False

    signals30 = {
        "embedding_model": re.compile(
            r"SentenceTransformer|multilingual-e5|all-MiniLM|"
            r"query:\s|passage:\s|embed_documents|embed_query",
            re.I,
        ),
        "vector_search": re.compile(
            r"\bfaiss\b|\bhnswlib\b|\bchromadb\b|\bqdrant\b|"
            r"cosine_similarity|semantic_search|IndexFlat|"
            r"nearest_neighbors",
            re.I,
        ),
        "embedding_compute": re.compile(
            r"\.encode\s*\(|normalize_embeddings|"
            r"mean_pooling|attention_mask.*pool",
            re.I,
        ),
    }

    excluded_parts30 = {
        ".git", ".venv", "venv", "rag_env", "__pycache__",
        ".ipynb_checkpoints", "backup_versions", "node_modules",
    }

    def within_budget30():
        return (
            scanned30 < max_files30
            and time.monotonic() - started30 < max_seconds30
        )

    print("بررسی کدهای پروژه؛ خروجی سل‌ها خوانده نمی‌شود…", flush=True)

    for folder_name30 in ("notebooks", "src", "scripts"):
        folder30 = root30 / folder_name30
        if not folder30.is_dir():
            continue

        for path30 in sorted(folder30.rglob("*")):
            if not within_budget30():
                stopped30 = True
                break
            if any(part in excluded_parts30 for part in path30.parts):
                continue
            if path30.is_symlink() or path30.suffix not in (".py", ".ipynb"):
                continue
            if not path30.is_file():
                continue

            scanned30 += 1
            relative30 = path30.relative_to(root30).as_posix()

            try:
                size30 = path30.stat().st_size
                if size30 > max_code_bytes30:
                    issues30.append({
                        "path": relative30,
                        "reason": "CODE_FILE_OVER_SIZE_LIMIT",
                    })
                    continue

                raw30 = path30.read_text(encoding="utf-8")
                if path30.suffix == ".ipynb":
                    notebook30 = json.loads(raw30)
                    units30 = [
                        (i, "".join(cell.get("source", [])))
                        for i, cell in enumerate(notebook30.get("cells", []))
                        if cell.get("cell_type") == "code"
                    ]
                else:
                    units30 = [(None, raw30)]

                matched30 = []
                for cell_index30, source30 in units30:
                    names30 = [
                        name for name, pattern in signals30.items()
                        if pattern.search(source30)
                    ]
                    if names30:
                        matched30.append({
                            "cell_index_zero_based": cell_index30,
                            "signals": names30,
                        })

                if matched30:
                    code_matches30.append({
                        "path": relative30,
                        "bytes": size30,
                        "matches": matched30,
                    })
            except (OSError, UnicodeError, ValueError) as error30:
                issues30.append({
                    "path": relative30,
                    "reason": type(error30).__name__,
                })

        if stopped30:
            break

    print("بررسی نام و اندازهٔ فایل‌های برداری؛ بدون بارگذاری…", flush=True)

    asset_folders30 = (
        "data/embeddings",
        "data/vectorstore",
        "data/index",
        "data/phase2/embeddings",
        "data/phase2/vectorstore",
        "data/phase2/index",
        "artifacts/phase2/embedding_benchmark",
    )

    checked_folders30 = []
    for relative_folder30 in asset_folders30:
        folder30 = root30 / relative_folder30
        checked_folders30.append({
            "path": relative_folder30,
            "exists": folder30.is_dir(),
        })
        if not folder30.is_dir():
            continue

        for path30 in sorted(folder30.rglob("*")):
            if not within_budget30():
                stopped30 = True
                break
            if path30.is_symlink() or not path30.is_file():
                continue

            scanned30 += 1
            try:
                assets30.append({
                    "path": path30.relative_to(root30).as_posix(),
                    "bytes": path30.stat().st_size,
                    "content_loaded": False,
                })
            except OSError as error30:
                issues30.append({
                    "path": path30.relative_to(root30).as_posix(),
                    "reason": type(error30).__name__,
                })

        if stopped30:
            break

    C030.update({
        "status": "PARTIAL_BOUNDED" if stopped30 else "COMPLETE_IN_SCOPE",
        "code_files_with_signals": len(code_matches30),
        "asset_files_found": len(assets30),
        "asset_total_bytes": sum(row["bytes"] for row in assets30),
        "code_matches": code_matches30,
        "asset_folders_checked": checked_folders30,
        "issues": issues30,
        "elapsed_seconds": time.monotonic() - started30,
        "scope": "LISTED_LOCAL_PATHS_AND_CODE_SIGNALS_ONLY",
        "runtime_or_integrity_verified": False,
    })

    C030["asset_inventory_artifact"] = save(
        "C030.semantic_asset_inventory", assets30, rows=True
    )
    C030["report"] = save("C030.semantic_inventory_report", C030.copy())

    print(json.dumps(C030, ensure_ascii=False, indent=2))

    print("\nبزرگ‌ترین فایل‌های شناسایی‌شده:")
    for row30 in sorted(
        assets30, key=lambda row: row["bytes"], reverse=True
    )[:15]:
        print(
            f"{row30['bytes'] / 1024**2:.2f} MiB | {row30['path']}"
        )

except Exception as exc:
    C030.update({"status": "FAILED", "error": str(exc)})
    print(json.dumps(C030, ensure_ascii=False, indent=2))
    raise
```

```python
# C031 v1 — Cached E5 vectors -> direct semantic rankings -> coverage.
# No model loading, embedding computation, Qdrant access or network.
# np.einsum(optimize=False) avoids a multi-threaded BLAS matrix product.

import io
import json
import math
import time
import hashlib
import numpy as np
from IPython.display import display, HTML

C031 = {
    "version": "NB22E_CACHED_DENSE_CANDIDATE_COVERAGE_V1",
    "status": "PREPARING",
    "new_model_pairs": 0,
    "new_query_embeddings": 0,
    "new_corpus_embeddings": 0,
    "new_api_calls": 0,
    "downloads": False,
    "automatic_replacement": False,
    "release_acceptance_granted": False,
}


def c031_receipt(path, raw):
    return {
        "path": path.relative_to(Path(ROOT).resolve()).as_posix(),
        "file_bytes_sha256": hashlib.sha256(raw).hexdigest(),
        "bytes": len(raw),
    }


def c031_summary(rows):
    require(bool(rows), "EMPTY_SUMMARY")
    result = {
        key: math.fsum(row[key] for row in rows) / len(rows)
        for key in (
            "top40_ceiling", "top100_ceiling",
            "semantic_selection_ceiling", "delta_vs_top100",
        )
    }
    result.update({
        "queries": len(rows),
        "queries_better_than_top100": sum(
            r["delta_vs_top100"] > 1e-12 for r in rows
        ),
        "queries_worse_than_top100": sum(
            r["delta_vs_top100"] < -1e-12 for r in rows
        ),
        "queries_unchanged_vs_top100": sum(
            abs(r["delta_vs_top100"]) <= 1e-12 for r in rows
        ),
        "oracle_target_capacity": (
            result["semantic_selection_ceiling"] >= 0.85
        ),
    })
    return result


try:
    started31 = time.perf_counter()
    require(C030.get("status") == "COMPLETE_IN_SCOPE", "C030_REQUIRED")

    # Exact hashes of the three files supplied in this conversation.
    handoff31 = read_json({
        "path": "artifacts/phase2/embedding/nb14_dense_index_handoff_v1.json",
        "sha256": "77eb9f34e36f174cf748fd99caa1d507b5dc8e4abcb51f62b452c35d46688efc",
    })
    query_meta31 = read_json({
        "path": (
            "artifacts/phase2/hybrid_retrieval/query_embedding_cache_v1/"
            "development_query_embeddings.json"
        ),
        "sha256": "bc97f189d5cadb801f5aaaad9a86715509ad2a91a18dc4840b0918c0c8f2568e",
    })
    manifest31 = read_json({
        "path": (
            "data/phase2/retrieval/hybrid_retrieval_v1/"
            "hybrid_retrieval_manifest.json"
        ),
        "sha256": "362b9cbfd33e2f8e307ffc6cc7dcffc48d2cc308fe74f9bb0ba896d1e080032c",
    })

    embedding31 = handoff31["embedding_config"]
    contract31 = query_meta31["contract"]
    require(
        embedding31["model_id"] == contract31["model_id"]
        == "intfloat/multilingual-e5-small"
        and embedding31["revision"] == contract31["model_revision"]
        == "614241f622f53c4eeff9890bdc4f31cfecc418b3"
        and embedding31["dimension"] == contract31["dimension"] == 384
        and embedding31["query_prefix"] == contract31["query_prefix"]
        == "query:"
        and embedding31["passage_prefix"] == "passage:"
        and contract31["normalize_embeddings"] is True
        and contract31["split"] == "development",
        "EMBEDDING_CONTRACT_CONFLICT",
    )
    require(
        query_meta31["cache_key"] == digest(canonical(contract31)),
        "QUERY_CACHE_CONTRACT_CHECKSUM_CONFLICT",
    )

    # Read the pinned chunk corpus; do not depend on a mutable global mapping.
    require(
        handoff31["chunk_input"]["sha256"]
        == manifest31["inputs"]["chunk"]["sha256"],
        "CHUNK_BINDING_CONFLICT",
    )
    chunk_rows31 = read_rows(handoff31["chunk_input"])
    chunks31 = keyed(chunk_rows31, "chunk_id")
    require(len(chunks31) == 3488, "CORPUS_COUNT_CONFLICT")

    # Verify Golden too, for exact question identity.
    golden_rows31 = read_rows(manifest31["inputs"]["golden_dataset"])
    golden31 = keyed(golden_rows31, "candidate_id")

    query_ids31 = query_meta31["candidate_ids"]
    require(
        len(query_ids31) == len(set(query_ids31)) == 2664
        and query_ids31 == sorted(query_ids31)
        and set(query_ids31) == set(LAB["dev"]["ids"]),
        "DEVELOPMENT_QUERY_IDS_CONFLICT",
    )

    q_digest31 = hashlib.sha256()
    for qid in query_ids31:
        component = (
            qid + "\x1f" + golden31[qid]["evaluation_content"]["question"]
        ).encode("utf-8")
        q_digest31.update(len(component).to_bytes(8, "big"))
        q_digest31.update(component)

    require(
        q_digest31.hexdigest() == contract31["query_set_sha256"],
        "QUERY_TEXT_BINDING_CONFLICT",
    )

    # Load query vectors only from the hash-bound existing cache.
    query_binding31 = manifest31["artifacts"]["development_query_embeddings"]
    require(
        query_binding31["path"] == query_meta31["embedding_path"]
        and binding_hash(query_binding31)
        == query_meta31["embedding_file_sha256"],
        "QUERY_VECTOR_BINDING_CONFLICT",
    )
    print("بررسی بردارهای سؤال و قطعه؛ بدون اجرای مدل…", flush=True)

    query_raw31 = checked_bytes(query_binding31)
    query_vectors31 = np.load(io.BytesIO(query_raw31), allow_pickle=False)
    require(
        query_vectors31.shape == (2664, 384)
        and query_vectors31.dtype == np.float32
        and np.isfinite(query_vectors31).all()
        and np.allclose(
            np.linalg.norm(query_vectors31, axis=1),
            1.0, rtol=1e-5, atol=1e-6,
        ),
        "INVALID_QUERY_VECTORS",
    )
    del query_raw31
    query_positions31 = {q: i for i, q in enumerate(query_ids31)}

    # Resolve the same passage-key contract used by NB14.
    chunk_ids31 = sorted(chunks31)
    passage_keys31 = {
        cid: "passage:" + hashlib.sha256(
            " ".join(str(chunks31[cid]["content"]).split()).encode("utf-8")
        ).hexdigest()
        for cid in chunk_ids31
    }
    needed_keys31 = set(passage_keys31.values())

    cache_root31 = path_for(handoff31["vector_cache"]["root"])
    require(cache_root31.is_dir(), "EXISTING_PASSAGE_CACHE_REQUIRED")

    release31 = read_json(handoff31["embedding_release"])
    npz_files31 = sorted(cache_root31.rglob("*.npz"))
    require(
        len(npz_files31) == release31["cache"]["npz_file_count"],
        "CORPUS_CACHE_FILE_COUNT_CONFLICT",
    )

    passage_vectors31 = {}
    seen_keys31 = set()
    asset_receipts31 = []
    total_cache_rows31 = 0

    for path in npz_files31:
        require(
            path.resolve().is_relative_to(Path(ROOT).resolve()),
            "CACHE_PATH_OUTSIDE_PROJECT",
        )
        raw = path.read_bytes()
        asset_receipts31.append(c031_receipt(path, raw))

        with np.load(io.BytesIO(raw), allow_pickle=False) as archive:
            keys = archive["input_keys"]
            vectors = archive["embeddings"]
            require(
                keys.ndim == 1
                and vectors.shape == (len(keys), 384)
                and vectors.dtype == np.float32
                and np.isfinite(vectors).all(),
                "INVALID_CORPUS_CACHE_ARRAY",
            )
            total_cache_rows31 += len(keys)

            for raw_key, vector in zip(keys, vectors):
                key = str(raw_key)
                require(key not in seen_keys31, "DUPLICATE_VECTOR_CACHE_KEY")
                seen_keys31.add(key)
                if key in needed_keys31:
                    passage_vectors31[key] = vector.copy()

        time.sleep(0.01)

    require(
        needed_keys31.issubset(passage_vectors31),
        "PASSAGE_VECTORS_MISSING_NO_RECOMPUTATION",
    )

    corpus_vectors31 = np.stack([
        passage_vectors31[passage_keys31[cid]] for cid in chunk_ids31
    ])
    require(
        corpus_vectors31.shape == (3488, 384)
        and np.max(np.abs(
            np.linalg.norm(corpus_vectors31, axis=1) - 1.0
        )) <= 1e-4,
        "CORPUS_VECTOR_NORMALIZATION_CONFLICT",
    )
    del passage_vectors31

    # Define one candidate selection before evaluating its coverage.
    specification31 = {
        "method": "BASELINE_TOP40_PLUS_DENSE60",
        "protected_baseline_prefix": 40,
        "maximum_candidates": 100,
        "dense_score": "DOT_PRODUCT_OF_EXISTING_L2_NORMALIZED_VECTORS",
        "dense_tie_break": "CHUNK_ID",
        "selection_uses_gold": False,
        "parameter_search": False,
        "model_id": embedding31["model_id"],
        "revision": embedding31["revision"],
        "original_corpus_inventory_hash_verified": False,
    }
    C031["specification_artifact"] = save(
        "C031.cached_dense_spec", specification31
    )
    C031["cache_receipts_artifact"] = save(
        "C031.corpus_cache_receipts", asset_receipts31, rows=True
    )

    coverage31 = read_json(C025["report"])
    reference_rows31 = read_rows(C025["cases_artifact"])
    reference31 = {}
    for row in reference_rows31:
        if row["candidate_method"] == "baseline_top100":
            require(row["query_id"] not in reference31, "DUPLICATE_REFERENCE")
            reference31[row["query_id"]] = row

    require(len(reference31) == 120, "EXPECTED_120_QUERIES")
    baseline_rows31 = c025_subset(
        coverage31["baseline_binding"], set(reference31)
    )

    candidates31 = {}
    dense_rankings31 = {}
    cases31 = []

    print("محاسبهٔ رتبه‌های معنایی از بردارهای موجود…", flush=True)

    for index, qid in enumerate(sorted(reference31), start=1):
        baseline = baseline_rows31[qid]["ranked_chunk_ids"]
        require(
            baseline_rows31[qid]["split"] == "development"
            and len(baseline) >= 100
            and len(baseline) == len(set(baseline)),
            "BASELINE_CONFLICT",
        )

        query_vector = query_vectors31[query_positions31[qid]]
        scores = np.einsum(
            "ij,j->i", corpus_vectors31, query_vector, optimize=False
        )
        require(np.isfinite(scores).all(), "NONFINITE_DENSE_SCORES")
        ranked = sorted(
            range(len(chunk_ids31)),
            key=lambda i: (-float(scores[i]), chunk_ids31[i]),
        )
        dense_order = [chunk_ids31[i] for i in ranked]

        selected = list(baseline[:40])
        selected_set = set(selected)
        for cid in dense_order:
            if cid not in selected_set:
                selected.append(cid)
                selected_set.add(cid)
            if len(selected) == 100:
                break

        require(
            len(selected) == len(set(selected)) == 100
            and selected[:40] == baseline[:40],
            "CANDIDATE_SELECTION_CONFLICT",
        )
        candidates31[qid] = selected
        dense_rankings31[qid] = dense_order[:100]

        # Gold is used only after candidate selection.
        gold = set(LAB["dev"]["gold"][qid])
        require(bool(gold), "EMPTY_GOLD")
        ceiling = lambda ids: min(10, len(set(ids) & gold)) / len(gold)
        top100 = ceiling(baseline[:100])
        require(
            abs(top100 - reference31[qid]["recall_at_10_ceiling"]) <= 1e-12,
            "TOP100_CEILING_REPLAY_CONFLICT",
        )

        cases31.append({
            "query_id": qid,
            "cohort": reference31[qid]["cohort"],
            "source_types": reference31[qid]["source_types"],
            "top40_ceiling": ceiling(baseline[:40]),
            "top100_ceiling": top100,
            "semantic_selection_ceiling": ceiling(selected),
            "delta_vs_top100": ceiling(selected) - top100,
        })

        if index % 20 == 0:
            print(json.dumps({
                "queries_completed": index,
                "queries_total": 120,
                "new_embeddings": 0,
                "new_model_pairs": 0,
            }), flush=True)
        time.sleep(0.05)

    cohorts31 = {}
    slices31 = {}
    for name in ("selection_dev60", "additional_dev60"):
        rows = [r for r in cases31 if r["cohort"] == name]
        require(len(rows) == 60, "COHORT_SIZE_CONFLICT")
        cohorts31[name] = c031_summary(rows)
        slices31[name] = {}
        for source in ("book", "transcript", "website"):
            source_rows = [r for r in rows if source in r["source_types"]]
            require(len(source_rows) == 20, "SOURCE_SIZE_CONFLICT")
            slices31[name][source] = c031_summary(source_rows)

    C031.update({
        "status": "COMPLETE",
        "cohorts": cohorts31,
        "per_source": slices31,
        "combined_descriptive_only": c031_summary(cases31),
        "corpus_chunks": 3488,
        "development_query_vectors": 2664,
        "cache_rows_read": total_cache_rows31,
        "original_corpus_inventory_hash_verified": False,
        "corpus_cache_validation_scope": (
            "EXACT_PASSAGE_KEYS_DIMENSION_DTYPE_FINITE_NORMALIZED;"
            "OBSERVED_FILE_HASHES_RECORDED_NOT_MATCHED_TO_OLD_INVENTORY"
        ),
        "qdrant_accessed": False,
        "passes_exploratory_capacity_screen": (
            all(r["delta_vs_top100"] >= -1e-12 for r in cohorts31.values())
            and any(r["delta_vs_top100"] > 1e-12 for r in cohorts31.values())
        ),
        "elapsed_seconds": time.perf_counter() - started31,
        "scope": "DEVELOPMENT_ORACLE_CAPACITY_ONLY_NOT_ACCEPTANCE",
    })

    C031["candidate_artifact"] = save("C031.cached_dense_candidates", {
        "specification": specification31,
        "candidates": candidates31,
        "dense_top100": dense_rankings31,
    })
    C031["cases_artifact"] = save(
        "C031.cached_dense_coverage_cases", cases31, rows=True
    )
    C031["report"] = save("C031.cached_dense_coverage_report", C031.copy())

    html_rows31 = []
    for name, label in (
        ("selection_dev60", "۶۰ سؤال اول"),
        ("additional_dev60", "۶۰ سؤال دوم"),
    ):
        groups = [(label, cohorts31[name])]
        for source, source_label in (
            ("book", "کتاب"),
            ("transcript", "متن صوتی"),
            ("website", "وب‌سایت"),
        ):
            groups.append((label + "؛ " + source_label, slices31[name][source]))

        for label, result in groups:
            tick = "✅" if result["oracle_target_capacity"] else ""
            html_rows31.append(
                f"<tr><td>{label}</td>"
                f"<td>{result['top40_ceiling']:.2%}</td>"
                f"<td>{result['top100_ceiling']:.2%}</td>"
                f"<td>{result['semantic_selection_ceiling']:.2%}</td>"
                f"<td>{100*result['delta_vs_top100']:+.2f}</td>"
                f"<td>{result['queries_better_than_top100']}</td>"
                f"<td>{result['queries_worse_than_top100']}</td>"
                f"<td>85%</td><td>{tick}</td></tr>"
            )

    display(HTML(
        '<div dir="rtl" style="text-align:right">'
        "<p><strong>سقف نظری Recall@10؛ مدل اجرا نشده است.</strong> "
        "تیک فقط ظرفیت نظری است، نه پذیرش. هدف منابع الزام مستقل نیست.</p>"
        '<table style="text-align:center">'
        "<tr><th>مجموعه</th><th>Top40</th><th>Top100</th>"
        "<th>Top40 + معنایی</th><th>تغییر نسبت به Top100؛ واحد درصد</th>"
        "<th>سؤال بهتر</th><th>سؤال بدتر</th><th>هدف</th>"
        "<th>ظرفیت کافی</th></tr>"
        + "".join(html_rows31) + "</table></div>"
    ))
    print(json.dumps(C031, ensure_ascii=False, indent=2))

except Exception as exc:
    C031.update({"status": "FAILED", "error": str(exc)})
    print(json.dumps(C031, ensure_ascii=False, indent=2))
    raise
```

```python
# C033 v1 — Common-question comparison using existing E/F artifacts.
# No BGE, retraining, embedding computation, network, or baseline mutation.

import json
import math
import hashlib
import time
import numpy as np
from IPython.display import display, HTML

C033 = {
    "version": "NB22E_COMMON_COHORT_COMPARISON_V1",
    "status": "PREPARING",
    "new_model_pairs": 0,
    "new_api_calls": 0,
    "downloads": False,
    "retraining": False,
    "automatic_replacement": False,
    "release_acceptance_granted": False,
}

KEYS33 = ("recall_at_10", "ndcg_at_10", "hit_at_10")


def c033_metric(order, gold):
    require(
        bool(gold) and len(order) == len(set(order)),
        "INVALID_METRIC_INPUT",
    )
    flags = [int(cid in gold) for cid in order[:10]]
    ideal = math.fsum(
        1 / math.log2(i + 2) for i in range(min(10, len(gold)))
    )
    return {
        "recall_at_10": sum(flags) / len(gold),
        "ndcg_at_10": math.fsum(
            hit / math.log2(i + 2) for i, hit in enumerate(flags)
        ) / ideal,
        "hit_at_10": float(any(flags)),
    }


def c033_mean(values):
    values = list(values)
    require(bool(values), "EMPTY_COHORT")
    return {
        key: math.fsum(row[key] for row in values) / len(values)
        for key in KEYS33
    }


def c033_check(actual, expected, label):
    require(
        all(abs(actual[k] - expected[k]) <= 1e-9 for k in KEYS33),
        "METRIC_REPLAY_CONFLICT:" + label,
    )


try:
    started33 = time.perf_counter()

    require(C027.get("status") == "COMPLETE", "C027_REQUIRED")
    require(C023.get("status") == "COMPLETE", "C023_REQUIRED")

    print("بررسی گزارش‌ها، مدل‌های کوچک و طرح تقسیم NB22F…", flush=True)

    # Exact final-report identity from the executed NB22F attachment.
    f_report_binding33 = {
        "path": (
            "artifacts/phase2/quality_improvement/NB22F.v1/"
            "F008.final_report."
            "05608869beb96f7f552a5850bcf5df3e6dd4ebeb7db5a431fb3fe5ff74accc2d.json"
        ),
        "sha256": (
            "05608869beb96f7f552a5850bcf5df3e6"
            "dd4ebeb7db5a431fb3fe5ff74accc2d"
        ),
    }

    f_report33 = read_json(f_report_binding33)
    f_plan33 = read_json(f_report33["plan"])
    f_models33 = read_json(f_report33["models"])

    require(
        f_report33["version"] == "NB22F.v1"
        and f_report33["status"] == "EXPERIMENT_COMPLETE",
        "NB22F_COMPLETED_REPORT_REQUIRED",
    )

    parts33 = f_plan33["partitions"]
    require(
        set(parts33) == {"fit", "validation", "check"},
        "NB22F_PARTITION_NAMES_CONFLICT",
    )
    role33 = {}
    for role, ids in parts33.items():
        for qid in ids:
            require(qid not in role33, "NB22F_PARTITION_OVERLAP")
            role33[qid] = role

    feature_id33 = f_models33["feature_id"]
    feature_folder33 = path_for(
        "artifacts/phase2/quality_improvement/NB22F.v1/"
        "features_" + feature_id33
    )
    require(feature_folder33.is_dir(), "EXISTING_NB22F_FEATURE_CACHE_REQUIRED")

    expected_features33 = [
        "dense_cosine", "baseline_rr", "word_rr", "char_rr",
        "hybrid_rr", "word_coverage", "numeric_coverage",
        "term_frequency", "log_length",
        "baseline_present", "hybrid_present",
    ]
    require(
        f_models33["features"] == expected_features33,
        "FEATURE_SCHEMA_CONFLICT",
    )

    mu33 = np.asarray(f_models33["mu"], dtype=np.float64)
    sd33 = np.asarray(f_models33["sd"], dtype=np.float64)
    ridge33 = np.asarray(f_models33["ridge"], dtype=np.float64)
    pair33 = np.asarray(f_models33["pairwise"], dtype=np.float64)

    require(
        mu33.shape == sd33.shape == pair33.shape == (11,)
        and ridge33.shape == (12,)
        and np.all(sd33 > 0)
        and all(np.isfinite(a).all() for a in (mu33, sd33, ridge33, pair33)),
        "LINEAR_MODEL_PARAMETER_CONFLICT",
    )

    # Load saved E results, not the E runtime variables.
    e40_report33 = read_json(C023["report"])
    e100_report33 = read_json(C027["report"])
    e40_orders33 = read_json(C023["orders_artifact"])
    e100_orders33 = read_json(C027["orders_artifact"])

    reference40_33 = {}
    cohorts33 = {}
    for row in e40_orders33["query_metrics"]:
        qid = row["query_id"]
        require(qid not in reference40_33, "DUPLICATE_E_QUERY")
        reference40_33[qid] = row
        cohorts33.setdefault(row["cohort"], []).append(qid)

    ids33 = sorted(reference40_33)
    require(
        len(ids33) == 120
        and set(ids33) == set(e100_orders33["orders"])
        and set(ids33).issubset(role33),
        "COMMON_QUERY_COVERAGE_CONFLICT",
    )
    require(
        all(len(ids) == 60 for ids in cohorts33.values())
        and len(cohorts33) == 2,
        "E_COHORT_SIZE_CONFLICT",
    )

    # Reuse the exact baseline and pool bound to the F experiment.
    baseline_rows33 = c025_subset(f_plan33["baseline"], set(ids33))
    pool_rows33 = c025_subset(f_plan33["pool"], set(ids33))

    route_names33 = (
        "baseline",
        "bge_rrf_top40",
        "bge_rrf_top100",
        "F_dense_only",
        "F_word_char_rrf",
        "F_dense_word_char_rrf",
        "F_linear_ridge",
        "F_pairwise_linear",
    )
    measured33 = {name: {} for name in route_names33}
    gaps33 = {name: {} for name in route_names33}
    query_rows33 = []

    print("مقایسهٔ هشت مسیر روی سؤال‌های یکسان؛ بدون اجرای مدل عصبی…", flush=True)

    for index33, qid in enumerate(ids33, start=1):
        baseline = baseline_rows33[qid]["ranked_chunk_ids"]
        pool = pool_rows33[qid]["candidate_chunk_ids"]

        require(
            baseline_rows33[qid]["split"]
            == pool_rows33[qid]["split"] == "development",
            "DEVELOPMENT_SPLIT_REQUIRED",
        )

        # This reproduces NB22F's candidate order exactly.
        f_candidates = list(dict.fromkeys(baseline[:40] + pool))
        cache_path = feature_folder33 / (
            hashlib.sha256(qid.encode()).hexdigest() + ".json"
        )
        require(cache_path.is_file(), "FEATURE_CACHE_MISSING:" + qid)
        cache = decode(cache_path.read_bytes())
        payload = cache["payload"]

        require(
            cache["feature_id"] == feature_id33
            and cache["query_id"] == qid
            and cache["checksum"] == digest(canonical(payload))
            and payload["ids"] == f_candidates,
            "FEATURE_CACHE_IDENTITY_CONFLICT:" + qid,
        )

        x = np.asarray(payload["x"], dtype=np.float64)
        require(
            x.shape == (len(f_candidates), 11)
            and np.isfinite(x).all(),
            "FEATURE_ARRAY_CONFLICT",
        )
        z = (x - mu33) / sd33

        # Element-wise sums avoid a multi-threaded BLAS operation.
        ridge_scores = ridge33[0] + np.sum(z * ridge33[1:], axis=1)
        pair_scores = np.sum(z * pair33, axis=1)

        def ranked(scores):
            return [
                f_candidates[i] for i in sorted(
                    range(len(f_candidates)),
                    key=lambda i: (-float(scores[i]), f_candidates[i]),
                )
            ]

        dense_order = ranked(x[:, 0])
        dense_ranks = {cid: i + 1 for i, cid in enumerate(dense_order)}
        dense_rr = np.asarray([
            1 / (60 + dense_ranks[cid]) for cid in f_candidates
        ])

        orders = {
            "baseline": baseline,
            "bge_rrf_top40": e40_orders33["orders"][qid],
            "bge_rrf_top100": e100_orders33["orders"][qid],
            "F_dense_only": dense_order,
            "F_word_char_rrf": ranked(x[:, 2] + x[:, 3]),
            "F_dense_word_char_rrf": ranked(dense_rr + x[:, 2] + x[:, 3]),
            "F_linear_ridge": ranked(ridge_scores),
            "F_pairwise_linear": ranked(pair_scores),
        }

        candidate_sets = {
            "baseline": set(baseline[:100]),
            "bge_rrf_top40": set(baseline[:40]),
            "bge_rrf_top100": set(baseline[:100]),
            **{
                name: set(f_candidates)
                for name in route_names33 if name.startswith("F_")
            },
        }

        gold = set(LAB["dev"]["gold"][qid])
        require(bool(gold), "EMPTY_GOLD")

        for name in route_names33:
            order = orders[name]
            candidates = candidate_sets[name]
            require(
                set(order[:10]).issubset(candidates),
                "OUTPUT_OUTSIDE_DECLARED_CANDIDATES",
            )
            value = c033_metric(order, gold)
            measured33[name][qid] = value

            available = len(candidates & gold)
            ceiling = min(10, available) / len(gold)
            outside_gap = len(gold - candidates) / len(gold)
            capacity_gap = max(0, available - 10) / len(gold)
            ranking_gap = ceiling - value["recall_at_10"]

            require(
                ranking_gap >= -1e-12
                and abs(
                    1 - value["recall_at_10"]
                    - outside_gap - capacity_gap - ranking_gap
                ) <= 1e-12,
                "GAP_DECOMPOSITION_CONFLICT",
            )
            gaps33[name][qid] = {
                "recall_ceiling": ceiling,
                "outside_candidate_gap": outside_gap,
                "top10_capacity_gap": capacity_gap,
                "ranking_gap": max(0.0, ranking_gap),
            }

        c033_check(
            measured33["baseline"][qid],
            reference40_33[qid]["before"],
            "baseline/" + qid,
        )
        c033_check(
            measured33["bge_rrf_top40"][qid],
            reference40_33[qid]["fusion"],
            "BGE40/" + qid,
        )
        c033_check(
            measured33["bge_rrf_top100"][qid],
            e100_orders33["after_metrics"][qid],
            "BGE100/" + qid,
        )

        query_rows33.append({
            "query_id": qid,
            "E_cohort": reference40_33[qid]["cohort"],
            "F_role": role33[qid],
            "source_types": golden[qid]["source_types"],
            "metrics": {name: measured33[name][qid] for name in route_names33},
            "gaps": {name: gaps33[name][qid] for name in route_names33},
        })

        if index33 % 20 == 0:
            print(json.dumps({
                "queries_completed": index33,
                "total": 120,
                "BGE_pairs": 0,
                "retraining": False,
            }), flush=True)
        time.sleep(0.01)

    # Full 120 is descriptive: some queries are F training data.
    # F_check intersection is separately reported; E methods have seen these
    # questions during previous development, so it is still not a fresh holdout.
    subsets33 = {
        "all120_descriptive": ids33,
        **{name: sorted(ids) for name, ids in cohorts33.items()},
        "common_F_check": [q for q in ids33 if role33[q] == "check"],
    }
    summaries33 = {}

    for subset_name, subset_ids in subsets33.items():
        if not subset_ids:
            summaries33[subset_name] = {"queries": 0, "routes": {}}
            continue

        before = c033_mean(measured33["baseline"][q] for q in subset_ids)
        results = {}

        for name in route_names33:
            after = c033_mean(measured33[name][q] for q in subset_ids)
            results[name] = {
                "before": before,
                "after": after,
                "delta_percentage_points": {
                    k: (
                        0.0 if abs(after[k] - before[k]) < 1e-12
                        else 100 * (after[k] - before[k])
                    )
                    for k in KEYS33
                },
                "gaps": {
                    k: math.fsum(gaps33[name][q][k] for q in subset_ids)
                    / len(subset_ids)
                    for k in gaps33[name][subset_ids[0]]
                },
                "queries_improved_ndcg": sum(
                    measured33[name][q]["ndcg_at_10"]
                    > measured33["baseline"][q]["ndcg_at_10"] + 1e-12
                    for q in subset_ids
                ),
                "queries_worsened_ndcg": sum(
                    measured33[name][q]["ndcg_at_10"]
                    < measured33["baseline"][q]["ndcg_at_10"] - 1e-12
                    for q in subset_ids
                ),
            }

        summaries33[subset_name] = {
            "queries": len(subset_ids),
            "F_roles": {
                role: sum(role33[q] == role for q in subset_ids)
                for role in ("fit", "validation", "check")
            },
            "routes": results,
        }

    C033.update({
        "status": "COMPLETE",
        "F_report": f_report_binding33,
        "E_top40_report": C023["report"],
        "E_top100_report": C027["report"],
        "summaries": summaries33,
        "scope": "COMMON_QUESTIONS_EXPLORATORY_ONLY",
        "comparison_limits": [
            "All120 includes F training and validation questions.",
            "F_check intersection is held out from F fitting/selection only.",
            "E methods used these questions in prior development.",
            "No independent release acceptance or automatic winner selection.",
        ],
        "elapsed_seconds": time.perf_counter() - started33,
    })

    C033["query_artifact"] = save(
        "C033.common_query_comparison", query_rows33, rows=True
    )
    C033["report"] = save("C033.common_cohort_report", C033.copy())

    # Main table: same 120 questions, explicitly descriptive.
    summary = summaries33["all120_descriptive"]
    html_rows = []

    for name in route_names33:
        result = summary["routes"][name]
        for key, label, target in (
            ("recall_at_10", "Recall@10", 0.85),
            ("ndcg_at_10", "nDCG@10", 0.75),
            ("hit_at_10", "Hit@10", None),
        ):
            threshold = "تعیین نشده" if target is None else f"{target:.0%}"
            tick = (
                "✅"
                if target is not None and result["after"][key] >= target
                else ""
            )
            html_rows.append(
                f"<tr><td>{name}</td><td>{label}</td>"
                f"<td>{result['before'][key]:.2%}</td>"
                f"<td>{result['after'][key]:.2%}</td>"
                f"<td>{result['delta_percentage_points'][key]:+.2f}</td>"
                f"<td>{threshold}</td><td>{tick}</td></tr>"
            )

    gap_rows = []
    for name in route_names33:
        gap = summary["routes"][name]["gaps"]
        gap_rows.append(
            f"<tr><td>{name}</td>"
            f"<td>{gap['recall_ceiling']:.2%}</td>"
            f"<td>{100*gap['outside_candidate_gap']:.2f}</td>"
            f"<td>{100*gap['ranking_gap']:.2f}</td>"
            f"<td>{100*gap['top10_capacity_gap']:.2f}</td></tr>"
        )

    display(HTML(
        '<div dir="rtl" style="text-align:right">'
        "<p><strong>مقایسهٔ توصیفی روی همان ۱۲۰ سؤال.</strong> "
        "بعضی سؤال‌ها در آموزش NB22F بوده‌اند؛ این جدول پذیرش نیست.</p>"
        "<p>نقش سؤال‌ها در NB22F: "
        + str(summary["F_roles"])
        + "</p>"
        '<table style="text-align:center">'
        "<tr><th>روش</th><th>معیار</th><th>قبل</th><th>بعد</th>"
        "<th>تغییر؛ واحد درصد</th><th>حد</th><th>قبول عددی</th></tr>"
        + "".join(html_rows)
        + "</table><br>"
        "<p>تفکیک فاصلهٔ Recall؛ فاصله‌ها بر حسب واحد درصدند.</p>"
        '<table style="text-align:center">'
        "<tr><th>روش</th><th>سقف نامزدها</th>"
        "<th>فاصلهٔ پوشش</th><th>فاصلهٔ رتبه‌بندی</th>"
        "<th>محدودیت Top10</th></tr>"
        + "".join(gap_rows)
        + "</table></div>"
    ))

    print(json.dumps(C033, ensure_ascii=False, indent=2))

except Exception as exc:
    C033.update({"status": "FAILED", "error": str(exc)})
    print(json.dumps(C033, ensure_ascii=False, indent=2))
    raise
```

<h3>C034 — نسخهٔ ۱: یادگیری ترکیب baseline و BGE از کش موجود</h3>
<ul>
<li>نامزدها: همان Top100 baseline؛ بدون گسترش pool.</li>
<li>ویژگی‌ها: رتبهٔ baseline، رتبه و امتیاز BGE و توافق آن‌ها.</li>
<li>آموزش دو مدل کوچک با تنظیمات ثابت روی ۶۰ سؤال fit.</li>
<li>انتخاب روی ۲۶ سؤال validation؛ بررسی فقط یک گزینه روی ۳۴ سؤال check.</li>
<li>شرط انتخاب: بهبود nDCG و عدم افت Recall و Hit نسبت به ترکیب ثابت Top40.</li>
<li>اگر گزینه‌ای واجد شرط نبود، ترکیب ثابت Top40 حفظ می‌شود.</li>
<li>سؤال‌ها قبلاً در توسعه دیده شده‌اند؛ check، holdout مستقل ریلیز نیست.</li>
<li>خارج‌بودن از Golden به معنی اثبات نامرتبط‌بودن نیست؛
مدل‌ها برچسب‌های ثبت‌شده را یاد می‌گیرند.</li>
<li>بدون اجرای دوبارهٔ BGE، دانلود، API یا تغییر baseline محصول.</li>
</ul>
</div>

```python
# C034 v1 — Learn baseline/BGE combination from existing Top100 scores.
# No neural inference, network, parameter sweep, or product replacement.

import json
import math
import hashlib
import time
import numpy as np
from IPython.display import display, HTML

C034 = {
    "version": "NB22E_CACHED_BGE_LEARNED_COMBINATION_V1",
    "status": "PREPARING",
    "new_model_pairs": 0,
    "new_api_calls": 0,
    "downloads": False,
    "automatic_replacement": False,
    "release_acceptance_granted": False,
}

K34 = ("recall_at_10", "ndcg_at_10", "hit_at_10")


def c034_metric(order, gold):
    require(bool(gold) and len(order) == len(set(order)), "INVALID_ORDER")
    flags = [int(cid in gold) for cid in order[:10]]
    ideal = math.fsum(
        1 / math.log2(i + 2) for i in range(min(10, len(gold)))
    )
    return {
        "recall_at_10": sum(flags) / len(gold),
        "ndcg_at_10": math.fsum(
            v / math.log2(i + 2) for i, v in enumerate(flags)
        ) / ideal,
        "hit_at_10": float(any(flags)),
    }


def c034_mean(rows):
    rows = list(rows)
    require(bool(rows), "EMPTY_COHORT")
    return {
        k: math.fsum(row[k] for row in rows) / len(rows)
        for k in K34
    }


def c034_pass(candidate, reference):
    return (
        candidate["ndcg_at_10"] > reference["ndcg_at_10"] + 1e-12
        and candidate["recall_at_10"] >= reference["recall_at_10"] - 1e-12
        and candidate["hit_at_10"] >= reference["hit_at_10"] - 1e-12
    )


try:
    require(C033.get("status") == "COMPLETE", "C033_REQUIRED")

    started34 = time.perf_counter()
    audit34 = read_json(C033["report"])
    query_rows34 = read_rows(C033["query_artifact"])
    frozen34 = read_json(C020["method_artifact"])
    top100_report34 = read_json(C027["report"])
    top100_orders34 = read_json(C027["orders_artifact"])
    top40_orders34 = read_json(C023["orders_artifact"])

    role34 = {}
    expected34 = {}
    for row in query_rows34:
        qid = row["query_id"]
        require(qid not in role34, "DUPLICATE_QUERY")
        role34[qid] = row["F_role"]
        expected34[qid] = row

    parts34 = {
        name: sorted(q for q in role34 if role34[q] == name)
        for name in ("fit", "validation", "check")
    }
    require(
        {k: len(v) for k, v in parts34.items()}
        == {"fit": 60, "validation": 26, "check": 34},
        "EXPECTED_60_26_34_PARTITION",
    )

    # Verify the roles against the original NB22F plan.
    f_report34 = read_json(audit34["F_report"])
    f_plan34 = read_json(f_report34["plan"])
    for name, ids in parts34.items():
        require(
            set(ids).issubset(f_plan34["partitions"][name]),
            "NB22F_PARTITION_BINDING_CONFLICT",
        )

    score_hash34 = frozen34["score_contract_sha256"]
    require(
        digest(canonical(frozen34["score_contract"])) == score_hash34
        and top100_report34["specification"]["score_contract_sha256"]
        == score_hash34,
        "BGE_SCORE_CONTRACT_CONFLICT",
    )

    cache_folder34 = path_for(
        "artifacts/phase2/quality_improvement/NB22E.bge_top20_cache_v1/"
        + score_hash34
    )
    require(cache_folder34.is_dir(), "EXISTING_BGE_CACHE_REQUIRED")

    d_final34 = read_json(PINS["D_final"])
    baseline_rows34 = c025_subset(
        d_final34["retained_rankings"], set(role34)
    )

    specification34 = {
        "method_family": "SUPERVISED_BASELINE_BGE_COMBINATION",
        "depth": 100,
        "features": [
            "baseline_rr100",
            "bge_rr100",
            "bge_query_zscore",
            "bge_query_minmax",
            "rank_agreement",
            "joint_rr",
        ],
        "ridge_lambda": 10.0,
        "positive_training_weight": 10.0,
        "pair_steps": 300,
        "pair_batch": 128,
        "pair_l2_to_equal_rank_anchor": 0.1,
        "pair_seed": 2234,
        "candidate_methods": ["weighted_ridge", "anchored_pairwise"],
        "reference": "FIXED_BGE_RRF_TOP40",
        "selection": "VALIDATION_ONLY_STRICT_NDCG_GAIN_NO_RECALL_HIT_LOSS",
        "parameter_search": False,
        "score_contract_sha256": score_hash34,
        "scope": "EXPOSED_DEVELOPMENT_NOT_INDEPENDENT_ACCEPTANCE",
    }
    C034["plan_artifact"] = save("C034.learned_combination_plan", {
        "specification": specification34,
        "partitions": parts34,
        "F_plan": f_report34["plan"],
        "reference_report": C023["report"],
        "score_source_report": C027["report"],
    })

    print("بررسی کش ۱۲۰۰۰ امتیاز و بازسازی ویژگی‌ها…", flush=True)

    X34 = {}
    Y34 = {}
    HEAD34 = {}
    BASE34 = {}
    REF34 = {}
    REF_ORDERS34 = {}
    fixed100_metrics34 = {}

    for index, qid in enumerate(sorted(role34), start=1):
        baseline = baseline_rows34[qid]["ranked_chunk_ids"]
        require(
            baseline_rows34[qid]["split"] == "development"
            and len(baseline) >= 100
            and len(baseline) == len(set(baseline)),
            "BASELINE_ORDER_CONFLICT",
        )
        head = list(baseline[:100])
        question = golden[qid]["evaluation_content"]["question"]
        raw_scores = []

        for cid in head:
            passage = LAB["dev"]["chunks"][cid]["content"]
            identity = {
                "contract_sha256": score_hash34,
                "query_id": qid,
                "chunk_id": cid,
                "question_sha256": hashlib.sha256(
                    question.encode("utf-8")
                ).hexdigest(),
                "passage_sha256": hashlib.sha256(
                    passage.encode("utf-8")
                ).hexdigest(),
            }
            path = cache_folder34 / (
                digest(canonical(identity)) + ".json"
            )
            require(
                path.is_file(),
                "CACHE_MISS_NO_INFERENCE:" + qid + "/" + cid,
            )
            record = decode(path.read_bytes())
            body = record["body"]
            require(
                body["identity"] == identity
                and record["body_sha256"] == digest(canonical(body)),
                "CACHE_CHECKSUM_OR_IDENTITY_CONFLICT",
            )
            score = float(body["score"])
            require(math.isfinite(score), "NONFINITE_SCORE")
            raw_scores.append(score)

        raw_scores = np.asarray(raw_scores, dtype=np.float64)
        positions = np.arange(1, 101, dtype=np.float64)
        bge_indices = sorted(
            range(100), key=lambda i: (-raw_scores[i], i)
        )
        bge_positions = np.empty(100, dtype=np.float64)
        for rank, index_bge in enumerate(bge_indices, start=1):
            bge_positions[index_bge] = rank

        base_rr = 100 / (60 + positions)
        bge_rr = 100 / (60 + bge_positions)
        std = max(float(raw_scores.std()), 1e-8)
        spread = max(float(np.ptp(raw_scores)), 1e-8)

        X34[qid] = np.column_stack([
            base_rr,
            bge_rr,
            np.clip((raw_scores - raw_scores.mean()) / std, -8, 8),
            (raw_scores - raw_scores.min()) / spread,
            1 - np.abs(positions - bge_positions) / 99,
            base_rr * bge_rr,
        ])

        gold = set(LAB["dev"]["gold"][qid])
        require(bool(gold), "EMPTY_GOLD")
        Y34[qid] = np.asarray(
            [float(cid in gold) for cid in head]
        )
        HEAD34[qid] = head
        BASE34[qid] = list(baseline)
        REF_ORDERS34[qid] = top40_orders34["orders"][qid]
        REF34[qid] = c034_metric(REF_ORDERS34[qid], gold)

        fixed_scores = 0.5 * base_rr + 0.5 * bge_rr
        fixed_head = [
            head[i] for i in sorted(
                range(100), key=lambda i: (-fixed_scores[i], i)
            )
        ]
        fixed_order = fixed_head + baseline[100:]

        require(
            fixed_order == top100_orders34["orders"][qid],
            "FIXED_TOP100_ORDER_REPLAY_CONFLICT:" + qid,
        )
        fixed100_metrics34[qid] = c034_metric(fixed_order, gold)

        for key in K34:
            require(
                abs(
                    REF34[qid][key]
                    - expected34[qid]["metrics"]["bge_rrf_top40"][key]
                ) <= 1e-9
                and abs(
                    fixed100_metrics34[qid][key]
                    - top100_orders34["after_metrics"][qid][key]
                ) <= 1e-9,
                "SAVED_METRIC_REPLAY_CONFLICT",
            )

        if index % 30 == 0:
            print(json.dumps({
                "queries_verified": index,
                "total": 120,
                "new_BGE_pairs": 0,
            }), flush=True)

    # ---------- Fit: no validation/check labels used here ----------
    train_x34 = np.vstack([X34[q] for q in parts34["fit"]])
    train_y34 = np.concatenate([Y34[q] for q in parts34["fit"]])
    require(
        train_y34.sum() > 0 and train_y34.sum() < len(train_y34),
        "TRAINING_REQUIRES_BOTH_LABELS",
    )

    mu34 = train_x34.mean(axis=0)
    sd34 = train_x34.std(axis=0)
    sd34[sd34 < 1e-8] = 1
    Z34 = {q: (X34[q] - mu34) / sd34 for q in role34}

    ztrain34 = np.vstack([Z34[q] for q in parts34["fit"]])
    design34 = np.column_stack([np.ones(len(ztrain34)), ztrain34])
    weights34 = np.where(train_y34 > 0, 10.0, 1.0)

    # Small 7x7 solve; matrix accumulation avoids large BLAS products.
    normal34 = np.zeros((7, 7), dtype=np.float64)
    rhs34 = np.zeros(7, dtype=np.float64)
    for row, target, weight in zip(design34, train_y34, weights34):
        normal34 += weight * np.outer(row, row)
        rhs34 += weight * row * target
    penalty34 = 10.0 * np.eye(7)
    penalty34[0, 0] = 0
    ridge34 = np.linalg.solve(normal34 + penalty34, rhs34)

    # Anchor is exactly an equal combination of the two reciprocal ranks.
    anchor34 = np.zeros(6)
    anchor34[0] = 0.5 * sd34[0]
    anchor34[1] = 0.5 * sd34[1]
    pair_w34 = anchor34.copy()

    pair_groups34 = []
    for qid in parts34["fit"]:
        positive = np.flatnonzero(Y34[qid] > 0)
        negative = np.flatnonzero(Y34[qid] == 0)
        if len(positive) and len(negative):
            pair_groups34.append((qid, positive, negative))

    require(bool(pair_groups34), "NO_VALID_TRAINING_PAIRS")
    rng34 = np.random.default_rng(2234)

    print("آموزش دو مدل کوچک روی fit؛ بدون آموزش یا اجرای BGE…", flush=True)

    for step in range(300):
        deltas = []
        for _ in range(128):
            qid, positive, negative = pair_groups34[
                int(rng34.integers(len(pair_groups34)))
            ]
            pi = int(rng34.choice(positive))
            ni = int(rng34.choice(negative))
            deltas.append(Z34[qid][pi] - Z34[qid][ni])

        deltas = np.asarray(deltas)
        margin = np.clip(np.sum(deltas * pair_w34, axis=1), -30, 30)
        gradient = -(
            deltas / (1 + np.exp(margin))[:, None]
        ).mean(axis=0) + 0.1 * (pair_w34 - anchor34)
        pair_w34 -= 0.03 / math.sqrt(1 + step / 100) * gradient

        if step % 50 == 0:
            time.sleep(0.02)

    require(
        np.isfinite(ridge34).all() and np.isfinite(pair_w34).all(),
        "NONFINITE_LEARNED_PARAMETERS",
    )
    C034["models_artifact"] = save("C034.learned_models", {
        "specification": specification34,
        "fit_ids": parts34["fit"],
        "mu": mu34.tolist(),
        "sd": sd34.tolist(),
        "ridge": ridge34.tolist(),
        "pairwise": pair_w34.tolist(),
        "anchor": anchor34.tolist(),
        "fit_queries_with_pairs": len(pair_groups34),
    })

    def c034_order(qid, method):
        if method == "fixed_top40":
            return REF_ORDERS34[qid]
        if method == "weighted_ridge":
            scores = ridge34[0] + np.sum(
                Z34[qid] * ridge34[1:], axis=1
            )
        elif method == "anchored_pairwise":
            scores = np.sum(Z34[qid] * pair_w34, axis=1)
        else:
            raise RuntimeError("UNKNOWN_METHOD")
        indices = sorted(
            range(100), key=lambda i: (-float(scores[i]), i)
        )
        return (
            [HEAD34[qid][i] for i in indices]
            + BASE34[qid][100:]
        )

    def c034_evaluate(ids, method):
        return {
            q: c034_metric(
                c034_order(q, method), LAB["dev"]["gold"][q]
            )
            for q in ids
        }

    # ---------- Validation: choose one or keep fixed Top40 ----------
    validation34 = {}
    validation_values34 = {}
    methods34 = ("fixed_top40", "weighted_ridge", "anchored_pairwise")

    for method in methods34:
        values = c034_evaluate(parts34["validation"], method)
        validation_values34[method] = values
        validation34[method] = c034_mean(values.values())

    reference_val34 = validation34["fixed_top40"]
    qualified34 = [
        method for method in methods34
        if method != "fixed_top40"
        and c034_pass(validation34[method], reference_val34)
    ]
    winner34 = (
        sorted(
            qualified34,
            key=lambda m: (-validation34[m]["ndcg_at_10"], m),
        )[0]
        if qualified34 else "fixed_top40"
    )

    C034["selection_artifact"] = save("C034.validation_selection", {
        "winner": winner34,
        "qualified": qualified34,
        "validation_metrics": validation34,
        "models": C034["models_artifact"],
        "check_used_for_selection": False,
    })

    # ---------- Check: evaluate only the frozen selected option ----------
    check_ids34 = parts34["check"]
    check_values34 = c034_evaluate(check_ids34, winner34)
    check_after34 = c034_mean(check_values34.values())
    check_reference34 = c034_mean(REF34[q] for q in check_ids34)
    check_baseline34 = c034_mean(
        c034_metric(BASE34[q], LAB["dev"]["gold"][q])
        for q in check_ids34
    )

    changes34 = {
        key: {
            "improved_vs_fixed_top40": sum(
                check_values34[q][key] > REF34[q][key] + 1e-12
                for q in check_ids34
            ),
            "worsened_vs_fixed_top40": sum(
                check_values34[q][key] < REF34[q][key] - 1e-12
                for q in check_ids34
            ),
        }
        for key in K34
    }

    slices34 = {}
    for source in ("book", "transcript", "website"):
        ids = [
            q for q in check_ids34
            if source in golden[q]["source_types"]
        ]
        slices34[source] = {
            "queries": len(ids),
            "fixed_top40": c034_mean(REF34[q] for q in ids) if ids else None,
            "selected": (
                c034_mean(check_values34[q] for q in ids) if ids else None
            ),
        }

    C034.update({
        "status": "COMPLETE",
        "specification": specification34,
        "partition_counts": {k: len(v) for k, v in parts34.items()},
        "validation_metrics": validation34,
        "selected_method": winner34,
        "check": {
            "queries": 34,
            "baseline": check_baseline34,
            "fixed_top40": check_reference34,
            "selected": check_after34,
            "strict_improvement_vs_fixed_top40": (
                winner34 != "fixed_top40"
                and c034_pass(check_after34, check_reference34)
            ),
            "absolute_targets_met": (
                check_after34["recall_at_10"] >= 0.85
                and check_after34["ndcg_at_10"] >= 0.75
            ),
            "query_changes": changes34,
            "per_source": slices34,
        },
        "scope": "EXPOSED_DEVELOPMENT_NOT_INDEPENDENT_ACCEPTANCE",
        "elapsed_seconds": time.perf_counter() - started34,
        "cached_BGE_pairs_consumed": 12000,
        "next_step": (
            "REVIEW_SINGLE_FROZEN_CHECK_RESULT;"
            "NO_CHECK_RUNNER_UP_SELECTION_OR_AUTOMATIC_RETUNING"
        ),
    })

    C034["check_artifact"] = save("C034.selected_check_results", {
        "selected_method": winner34,
        "orders": {
            q: c034_order(q, winner34) for q in check_ids34
        },
        "metrics": check_values34,
    })
    C034["report"] = save("C034.learned_combination_report", C034.copy())

    table_rows34 = []

    for section, method, before, after in (
        *[
            (
                "validation؛ ۲۶ سؤال",
                method,
                reference_val34,
                validation34[method],
            )
            for method in methods34
        ],
        (
            "check؛ ۳۴ سؤال",
            winner34,
            check_reference34,
            check_after34,
        ),
    ):
        for key, label, target in (
            ("recall_at_10", "Recall@10", 0.85),
            ("ndcg_at_10", "nDCG@10", 0.75),
            ("hit_at_10", "Hit@10", None),
        ):
            threshold = "تعیین نشده" if target is None else f"{target:.0%}"
            tick = "✅" if target is not None and after[key] >= target else ""
            delta = 100 * (after[key] - before[key])
            if abs(delta) < 1e-10:
                delta = 0.0

            table_rows34.append(
                f"<tr><td>{section}</td><td>{method}</td><td>{label}</td>"
                f"<td>{before[key]:.2%}</td><td>{after[key]:.2%}</td>"
                f"<td>{delta:+.2f}</td><td>{threshold}</td><td>{tick}</td></tr>"
            )

    display(HTML(
        '<div dir="rtl" style="text-align:right">'
        "<p><strong>مرجع ستون «قبل»، ترکیب ثابت BGE Top40 است.</strong> "
        "تنها یک گزینهٔ منتخب روی check بررسی شد. "
        "این سؤال‌ها قبلاً دیده شده‌اند؛ تیک پذیرش ریلیز نیست.</p>"
        '<table style="text-align:center">'
        "<tr><th>بخش</th><th>روش</th><th>معیار</th><th>قبل</th>"
        "<th>بعد</th><th>تغییر؛ واحد درصد</th>"
        "<th>حد پذیرش</th><th>قبول عددی</th></tr>"
        + "".join(table_rows34) + "</table></div>"
    ))

    print(json.dumps(C034, ensure_ascii=False, indent=2))

except Exception as exc:
    C034.update({"status": "FAILED", "error": str(exc)})
    print(json.dumps(C034, ensure_ascii=False, indent=2))
    raise
```

```python
# C035 v2 — Complete window-retrieval experiment with resumable caching.
# No BGE, embedding computation, network, or source-artifact modification.

import json
import math
import time
import hashlib
import collections
import unicodedata
import re
from IPython.display import display, HTML

C035 = {
    "version": "NB22E_WINDOW_RETRIEVAL_V2",
    "status": "PREPARING",
    "new_model_pairs": 0,
    "new_api_calls": 0,
    "downloads": False,
    "automatic_replacement": False,
    "release_acceptance_granted": False,
}

SESSION_SECONDS35 = 720



def c035_words(text):
    text = unicodedata.normalize("NFKC", text).lower()
    text = text.translate(str.maketrans(
        "يك٠١٢٣٤٥٦٧٨٩۰۱۲۳۴۵۶۷۸۹",
        "یک01234567890123456789",
    ))
    text = "".join(
        c for c in text
        if not unicodedata.combining(c) and c != "ـ"
    ).replace("\u200c", " ")
    return re.findall(r"[^\W_]+", text, flags=re.UNICODE)


def c035_metric(order, gold):
    require(
        bool(gold) and len(order) == len(set(order)),
        "INVALID_METRIC_INPUT",
    )
    flags = [int(cid in gold) for cid in order[:10]]
    ideal = math.fsum(
        1 / math.log2(i + 2)
        for i in range(min(10, len(gold)))
    )
    return {
        "recall_at_10": sum(flags) / len(gold),
        "ndcg_at_10": math.fsum(
            v / math.log2(i + 2) for i, v in enumerate(flags)
        ) / ideal,
        "hit_at_10": float(any(flags)),
    }


def c035_mean(values):
    values = list(values)
    require(bool(values), "EMPTY_COHORT")
    return {
        k: math.fsum(v[k] for v in values) / len(values)
        for k in ("recall_at_10", "ndcg_at_10", "hit_at_10")
    }


def c035_guard(candidate, reference):
    return (
        candidate["ndcg_at_10"] > reference["ndcg_at_10"] + 1e-12
        and candidate["recall_at_10"]
        >= reference["recall_at_10"] - 1e-12
        and candidate["hit_at_10"]
        >= reference["hit_at_10"] - 1e-12
    )


def c035_table(validation, winner, before=None, after=None):
    displayed = [
        ("validation؛ ۵۳۳ سؤال", route, validation["baseline"], values)
        for route, values in validation.items()
    ]
    if before is not None and after is not None:
        displayed.append(("check؛ ۶۲۴ سؤال", winner, before, after))

    html_rows = []
    for section, route, reference, measured in displayed:
        for key, label, target in (
            ("recall_at_10", "Recall@10", 0.85),
            ("ndcg_at_10", "nDCG@10", 0.75),
            ("hit_at_10", "Hit@10", None),
        ):
            tick = (
                "✅"
                if target is not None and measured[key] >= target
                else ""
            )
            threshold = (
                "تعیین نشده" if target is None else f"{target:.0%}"
            )
            delta = 100 * (measured[key] - reference[key])
            if abs(delta) < 1e-10:
                delta = 0.0

            html_rows.append(
                f"<tr><td>{section}</td><td>{route}</td>"
                f"<td>{label}</td><td>{reference[key]:.2%}</td>"
                f"<td>{measured[key]:.2%}</td><td>{delta:+.2f}</td>"
                f"<td>{threshold}</td><td>{tick}</td></tr>"
            )

    display(HTML(
        '<div dir="rtl" style="text-align:right">'
        "<p>مقایسه با baseline اصلی روی جمعیت یکسان. "
        "تیک عددی، پذیرش مستقل ریلیز نیست.</p>"
        '<table style="text-align:center">'
        "<tr><th>بخش</th><th>روش</th><th>معیار</th><th>قبل</th>"
        "<th>بعد</th><th>تغییر؛ واحد درصد</th>"
        "<th>حد پذیرش</th><th>قبول عددی</th></tr>"
        + "".join(html_rows) + "</table></div>"
    ))


try:
    require(C033.get("status") == "COMPLETE", "C033_REQUIRED")

    audit35 = read_json(C033["report"])
    f_report35 = read_json(audit35["F_report"])
    f_plan35 = read_json(f_report35["plan"])
    validation_ids35 = list(f_plan35["partitions"]["validation"])
    check_ids35 = list(f_plan35["partitions"]["check"])

    require(
        len(validation_ids35) == len(set(validation_ids35)) == 533
        and len(check_ids35) == len(set(check_ids35)) == 624
        and not set(validation_ids35).intersection(check_ids35),
        "FROZEN_PARTITION_CONFLICT",
    )

    print("بررسی فایل‌های اصلی و baseline…", flush=True)

    verified35 = read_json(PINS["C056"])
    chunk_binding35 = verified35["input_file_receipts"]["chunks"]
    golden_binding35 = verified35["input_file_receipts"]["golden"]

    chunks35 = keyed(read_rows(chunk_binding35), "chunk_id")
    golden35 = keyed(read_rows(golden_binding35), "candidate_id")
    needed35 = set(validation_ids35 + check_ids35)
    baseline_rows35 = c025_subset(f_plan35["baseline"], needed35)

    require(
        len(chunks35) == 3488
        and needed35.issubset(LAB["dev"]["gold"])
        and needed35.issubset(golden35),
        "INPUT_POPULATION_CONFLICT",
    )

    # Validate every baseline before any new retrieval.
    for qid in sorted(needed35):
        row = baseline_rows35[qid]
        order = row["ranked_chunk_ids"]

        require(
            row["split"] == "development"
            and len(order) >= 10
            and len(order) == len(set(order))
            and set(order).issubset(chunks35),
            "BASELINE_ORDER_CONFLICT:" + qid,
        )

        replay = c035_metric(order, LAB["dev"]["gold"][qid])
        require(
            all(
                abs(replay[k] - row["metrics"][k]) <= 1e-9
                for k in replay
            ),
            "BASELINE_METRIC_REPLAY_CONFLICT:" + qid,
        )

    spec35 = {
        "method_family": "WITHIN_CHUNK_WINDOW_BM25",
        "window_words": 90,
        "stride_words": 45,
        "bm25_k1": 1.2,
        "bm25_b": 0.75,
        "query_term_frequency": True,
        "chunk_aggregation": "MAX_WINDOW_SCORE",
        "positive_score_chunks_only": True,
        "maximum_window_rank_depth": 800,
        "fusion": "EQUAL_RANK_RRF_K60_WITH_FULL_BASELINE_ORDER",
        "routes": ["baseline", "window_only", "window_baseline_rrf"],
        "selection": "VALIDATION_STRICT_NDCG_GAIN_NO_RECALL_HIT_LOSS",
        "source_chunks_unchanged": True,
        "gold_used_for_retrieval": False,
        "parameter_search": False,
    }

    # Retrieval contract is unchanged from C035 v1, allowing cache reuse.
    identity35 = {
        "specification": spec35,
        "chunks": chunk_binding35,
        "golden": golden_binding35,
        "baseline": f_plan35["baseline"],
    }
    contract_hash35 = digest(canonical(identity35))
    cache_dir35 = path_for(
        "artifacts/phase2/quality_improvement/"
        "NB22E.window_retrieval_cache_v1/" + contract_hash35
    )
    cache_dir35.mkdir(parents=True, exist_ok=True)

    C035["plan_artifact"] = save("C035.window_retrieval_plan", {
        "identity": identity35,
        "contract_sha256": contract_hash35,
        "partitions": {
            "validation": validation_ids35,
            "check": check_ids35,
        },
        "session_seconds": SESSION_SECONDS35,
        "scope": "PREVIOUSLY_EXPOSED_DEVELOPMENT_NOT_INDEPENDENT_ACCEPTANCE",
    })

    # ---------- Build or reuse the in-memory window index ----------
    runtime_key35 = digest(canonical({
        "contract": contract_hash35,
        "implementation": "WINDOW_INDEX_V1",
    }))
    runtime35 = globals().get("WINDOW_INDEX_RUNTIME")

    if (
        not isinstance(runtime35, dict)
        or runtime35.get("key") != runtime_key35
    ):
        print("ساخت ایندکس پنجره‌ای؛ بدون تغییر متن اصلی…", flush=True)

        postings35 = collections.defaultdict(list)
        lengths35 = []
        owners35 = []

        for cid in sorted(chunks35):
            tokens = c035_words(chunks35[cid]["content"])
            require(bool(tokens), "EMPTY_TOKENIZED_CHUNK:" + cid)

            for start in range(0, len(tokens), 45):
                window = tokens[start:start + 90]
                wid = len(owners35)
                owners35.append(cid)
                lengths35.append(len(window))

                for term, tf in collections.Counter(window).items():
                    postings35[term].append((wid, tf))

                if start + 90 >= len(tokens):
                    break

        require(bool(owners35), "EMPTY_WINDOW_INDEX")
        runtime35 = {
            "key": runtime_key35,
            "postings": dict(postings35),
            "lengths": lengths35,
            "owners": owners35,
            "avgdl": math.fsum(lengths35) / len(lengths35),
        }
        WINDOW_INDEX_RUNTIME = runtime35

    print(json.dumps({
        "stage": "WINDOW_INDEX_READY",
        "chunks": len(chunks35),
        "windows": len(runtime35["owners"]),
        "session_minutes": SESSION_SECONDS35 / 60,
        "new_model_pairs": 0,
    }), flush=True)

    def c035_search(question):
        scores = {}
        n = len(runtime35["owners"])
        avgdl = runtime35["avgdl"]

        for term, qtf in collections.Counter(c035_words(question)).items():
            posting = runtime35["postings"].get(term, ())
            if not posting:
                continue

            df = len(posting)
            idf = math.log(1 + (n - df + 0.5) / (df + 0.5))

            for wid, tf in posting:
                norm = (
                    1 - 0.75
                    + 0.75 * runtime35["lengths"][wid] / avgdl
                )
                value = qtf * idf * tf * 2.2 / (tf + 1.2 * norm)
                scores[wid] = scores.get(wid, 0.0) + value

        chunk_scores = {}
        for wid, value in scores.items():
            cid = runtime35["owners"][wid]
            if value > chunk_scores.get(cid, 0.0):
                chunk_scores[cid] = value

        return sorted(
            chunk_scores,
            key=lambda cid: (-chunk_scores[cid], cid),
        )[:800]

    def c035_orders(qid, window_order):
        baseline = baseline_rows35[qid]["ranked_chunk_ids"]
        baseline_rank = {
            cid: i + 1 for i, cid in enumerate(baseline)
        }
        window_rank = {
            cid: i + 1 for i, cid in enumerate(window_order)
        }
        union = set(baseline_rank) | set(window_rank)

        def rrf(cid):
            return (
                (0.5 / (60 + baseline_rank[cid])
                 if cid in baseline_rank else 0.0)
                + (0.5 / (60 + window_rank[cid])
                   if cid in window_rank else 0.0)
            )

        fused = sorted(
            union,
            key=lambda cid: (
                -rrf(cid),
                baseline_rank.get(cid, len(baseline) + 1),
                cid,
            ),
        )
        return {
            "baseline": baseline,
            "window_only": window_order,
            "window_baseline_rrf": fused,
        }

    def c035_cached_order(qid):
        question = golden35[qid]["evaluation_content"]["question"]
        identity = {
            "contract_sha256": contract_hash35,
            "query_id": qid,
            "question_sha256": hashlib.sha256(question.encode()).hexdigest(),
        }
        path = cache_dir35 / (digest(canonical(identity)) + ".json")

        if path.is_file():
            record = decode(path.read_bytes())
            require(
                record["identity"] == identity
                and record["order_sha256"]
                == digest(canonical(record["order"])),
                "WINDOW_CACHE_CONFLICT",
            )
            order = record["order"]
            hit = True
        else:
            order = c035_search(question)
            record = {
                "identity": identity,
                "order": order,
                "order_sha256": digest(canonical(order)),
            }
            c017_atomic_json(path, record)
            hit = False

        require(
            len(order) <= 800
            and len(order) == len(set(order))
            and set(order).issubset(chunks35),
            "WINDOW_ORDER_CONFLICT",
        )
        return order, hit

    # ---------- Validation ----------
    start_session35 = time.monotonic()
    fresh35 = 0
    reused35 = 0
    routes35 = spec35["routes"]
    val_values35 = {route: {} for route in routes35}
    partial35 = False

    for index, qid in enumerate(validation_ids35, start=1):
        if time.monotonic() - start_session35 >= SESSION_SECONDS35:
            partial35 = True
            break

        window_order, hit = c035_cached_order(qid)
        reused35 += int(hit)
        fresh35 += int(not hit)

        orders = c035_orders(qid, window_order)
        gold = LAB["dev"]["gold"][qid]
        for route in routes35:
            val_values35[route][qid] = c035_metric(orders[route], gold)

        if index % 50 == 0:
            elapsed = time.monotonic() - start_session35
            print(json.dumps({
                "stage": "VALIDATION",
                "completed": index,
                "total": len(validation_ids35),
                "fresh_retrievals": fresh35,
                "cached_retrievals": reused35,
                "elapsed_minutes": round(elapsed / 60, 2),
            }), flush=True)
        time.sleep(0.01)

    if partial35:
        C035.update({
            "status": "PARTIAL_RESUMABLE",
            "stage": "validation",
            "message": "برای ادامه فقط همین سل را دوباره اجرا کن.",
        })

    else:
        validation35 = {
            route: c035_mean(values.values())
            for route, values in val_values35.items()
        }
        reference35 = validation35["baseline"]

        qualified35 = [
            route for route in routes35
            if route != "baseline"
            and c035_guard(validation35[route], reference35)
        ]
        winner35 = (
            sorted(
                qualified35,
                key=lambda route: (
                    -validation35[route]["ndcg_at_10"], route
                ),
            )[0]
            if qualified35 else "baseline"
        )

        C035["selection_artifact"] = save("C035.validation_selection", {
            "selected_method": winner35,
            "validation_metrics": validation35,
            "qualified": qualified35,
            "plan": C035["plan_artifact"],
            "check_used_for_selection": False,
        })

        print(json.dumps({
            "stage": "METHOD_FROZEN_BEFORE_CHECK",
            "selected_method": winner35,
            "qualified": qualified35,
        }), flush=True)

        # ---------- Check: one frozen method only ----------
        check_before35 = {}
        check_after35 = {}
        check_orders35 = {}

        for index, qid in enumerate(check_ids35, start=1):
            if time.monotonic() - start_session35 >= SESSION_SECONDS35:
                partial35 = True
                break

            baseline = baseline_rows35[qid]["ranked_chunk_ids"]

            if winner35 == "baseline":
                selected = baseline
            else:
                window_order, hit = c035_cached_order(qid)
                reused35 += int(hit)
                fresh35 += int(not hit)
                selected = c035_orders(qid, window_order)[winner35]

            gold = LAB["dev"]["gold"][qid]
            check_before35[qid] = c035_metric(baseline, gold)
            check_after35[qid] = c035_metric(selected, gold)
            check_orders35[qid] = selected

            if index % 50 == 0:
                print(json.dumps({
                    "stage": "FROZEN_CHECK",
                    "completed": index,
                    "total": len(check_ids35),
                    "selected_method": winner35,
                    "elapsed_minutes": round(
                        (time.monotonic() - start_session35) / 60, 2
                    ),
                }), flush=True)
            time.sleep(0.01)

        if partial35:
            C035.update({
                "status": "PARTIAL_RESUMABLE",
                "stage": "check",
                "selected_method": winner35,
                "message": "کش حفظ شده؛ فقط همین سل را دوباره اجرا کن.",
            })
            c035_table(validation35, winner35)

        else:
            before35 = c035_mean(check_before35.values())
            after35 = c035_mean(check_after35.values())
            source_slices35 = {}

            for source in ("book", "transcript", "website"):
                ids = [
                    q for q in check_ids35
                    if source in golden35[q]["source_types"]
                ]
                source_slices35[source] = {
                    "queries": len(ids),
                    "before": (
                        c035_mean(check_before35[q] for q in ids)
                        if ids else None
                    ),
                    "after": (
                        c035_mean(check_after35[q] for q in ids)
                        if ids else None
                    ),
                }

            changes35 = {}
            for key in ("recall_at_10", "ndcg_at_10", "hit_at_10"):
                deltas = [
                    check_after35[q][key] - check_before35[q][key]
                    for q in check_ids35
                ]
                changes35[key] = {
                    "improved": sum(d > 1e-12 for d in deltas),
                    "worsened": sum(d < -1e-12 for d in deltas),
                    "unchanged": sum(abs(d) <= 1e-12 for d in deltas),
                }

            C035.update({
                "status": "COMPLETE",
                "selected_method": winner35,
                "validation_queries": len(validation_ids35),
                "check_queries": len(check_ids35),
                "validation_metrics": validation35,
                "before": before35,
                "after": after35,
                "delta_percentage_points": {
                    key: (
                        0.0
                        if abs(after35[key] - before35[key]) <= 1e-12
                        else 100 * (after35[key] - before35[key])
                    )
                    for key in before35
                },
                "per_source": source_slices35,
                "query_changes": changes35,
                "strict_check_improvement": (
                    winner35 != "baseline"
                    and c035_guard(after35, before35)
                ),
                "absolute_targets_met": (
                    after35["recall_at_10"] >= 0.85
                    and after35["ndcg_at_10"] >= 0.75
                ),
                "scope": (
                    "EXPOSED_COMPONENT_SPLIT_DEVELOPMENT"
                    "_NOT_INDEPENDENT_RELEASE_ACCEPTANCE"
                ),
                "no_check_runner_up_selection": True,
                "execution_mode": "SEQUENTIAL_PYTHON_NO_NEURAL_RUNTIME",
            })

            C035["check_artifact"] = save("C035.check_orders", {
                "selected_method": winner35,
                "orders": check_orders35,
                "before_metrics": check_before35,
                "after_metrics": check_after35,
            })

            c035_table(
                validation35, winner35, before35, after35
            )

    C035["session_seconds"] = time.monotonic() - start_session35
    C035["fresh_lexical_queries_this_session"] = fresh35
    C035["cached_lexical_queries_this_session"] = reused35
    C035["report"] = save("C035.window_retrieval_report", C035.copy())

    print(json.dumps(C035, ensure_ascii=False, indent=2))

except KeyboardInterrupt:
    C035.update({
        "status": "INTERRUPTED_RESUMABLE",
        "message": "کش‌های کامل حفظ شده‌اند؛ همین سل را دوباره اجرا کن.",
    })
    print(json.dumps(C035, ensure_ascii=False, indent=2))
    raise

except Exception as exc:
    C035.update({"status": "FAILED", "error": str(exc)})
    print(json.dumps(C035, ensure_ascii=False, indent=2))
    raise
```

```python
# C036 v1 — Coverage ceiling, validated cache counts, and runtime estimate.
# No model loading, inference, downloads, cache writes, or baseline changes.

import json
import math
import time
import hashlib
from pathlib import Path
from IPython.display import display, HTML

C036 = {
    "version": "NB22E_LARGE_COHORT_BGE_FEASIBILITY_V1",
    "status": "PREPARING",
    "new_model_pairs": 0,
    "new_api_calls": 0,
    "downloads": False,
    "automatic_replacement": False,
    "release_acceptance_granted": False,
}


def c036_mean(rows):
    rows = list(rows)
    require(bool(rows), "EMPTY_COHORT")
    return {
        k: math.fsum(row[k] for row in rows) / len(rows)
        for k in ("recall_at_10", "ndcg_at_10", "hit_at_10")
    }


def c036_metric(order, gold):
    require(
        bool(gold) and len(order) == len(set(order)),
        "INVALID_METRIC_INPUT",
    )
    flags = [int(cid in gold) for cid in order[:10]]
    ideal = math.fsum(
        1 / math.log2(i + 2)
        for i in range(min(10, len(gold)))
    )
    return {
        "recall_at_10": sum(flags) / len(gold),
        "ndcg_at_10": math.fsum(
            value / math.log2(i + 2)
            for i, value in enumerate(flags)
        ) / ideal,
        "hit_at_10": float(any(flags)),
    }


try:
    started36 = time.perf_counter()
    require(C035.get("status") == "COMPLETE", "C035_REQUIRED")

    report35_36 = read_json(C035["report"])
    plan35_36 = read_json(C035["plan_artifact"])
    selected36 = list(plan35_36["partitions"]["validation"])

    require(
        len(selected36) == len(set(selected36)) == 533,
        "EXPECTED_533_VALIDATION_QUERIES",
    )

    baseline_binding36 = plan35_36["identity"]["baseline"]
    baseline_rows36 = c025_subset(
        baseline_binding36, set(selected36)
    )

    # Read exact source files instead of guessing their in-memory structure.
    verified36 = read_json(PINS["C056"])
    bindings36 = verified36["input_file_receipts"]
    golden36 = keyed(read_rows(bindings36["golden"]), "candidate_id")
    chunks36 = keyed(read_rows(bindings36["chunks"]), "chunk_id")

    frozen36 = read_json(C020["method_artifact"])
    score_spec36 = frozen36["score_contract"]
    score_hash36 = frozen36["score_contract_sha256"]

    require(
        digest(canonical(score_spec36)) == score_hash36
        and score_spec36["model_id"] == "BAAI/bge-reranker-v2-m3"
        and score_spec36["revision"]
        == "953dc6f6f85a1b2dbfca4c34a2796e7dde08d41e"
        and score_spec36["batch_size"] == 1
        and score_spec36["max_length"] == 512,
        "BGE_SCORE_CONTRACT_CONFLICT",
    )

    cache_folder36 = path_for(
        "artifacts/phase2/quality_improvement/"
        "NB22E.bge_top20_cache_v1/" + score_hash36
    )

    # A different runtime contract must not silently consume this cache.
    cache_threads36 = score_spec36["threads"]

    timing_report36 = read_json(C027["report"])
    require(
        timing_report36["specification"]["score_contract_sha256"]
        == score_hash36,
        "TIMING_SCORE_CONTRACT_CONFLICT",
    )
    timing_pairs36 = timing_report36["new_model_pairs"]
    timing_seconds36 = timing_report36["wall_seconds"]

    require(
        timing_pairs36 > 0 and timing_seconds36 > 0,
        "VALID_TIMING_REFERENCE_REQUIRED",
    )
    seconds_per_pair36 = timing_seconds36 / timing_pairs36

    print(
        "بررسی سقف پوشش و کش امتیازها؛ هیچ امتیاز تازه‌ای ساخته نمی‌شود…",
        flush=True,
    )

    cases36 = []
    baseline_metrics36 = {}
    counters36 = {
        40: {"valid_cached_pairs": 0, "missing_pairs": 0},
        100: {"valid_cached_pairs": 0, "missing_pairs": 0},
    }

    for index36, qid in enumerate(selected36, start=1):
        row36 = baseline_rows36[qid]
        order36 = row36["ranked_chunk_ids"]

        require(
            row36["split"] == "development"
            and len(order36) >= 100
            and len(order36) == len(set(order36))
            and set(order36).issubset(chunks36),
            "BASELINE_ORDER_CONFLICT:" + qid,
        )

        gold36 = set(LAB["dev"]["gold"][qid])
        measured36 = c036_metric(order36, gold36)

        require(
            all(
                abs(measured36[k] - row36["metrics"][k]) <= 1e-9
                for k in measured36
            ),
            "BASELINE_METRIC_REPLAY_CONFLICT:" + qid,
        )
        baseline_metrics36[qid] = measured36

        question36 = golden36[qid]["evaluation_content"]["question"]
        require(
            isinstance(question36, str) and question36.strip(),
            "INVALID_QUESTION:" + qid,
        )
        question_hash36 = hashlib.sha256(
            question36.encode("utf-8")
        ).hexdigest()

        cached_flags36 = []

        for cid36 in order36[:100]:
            passage36 = chunks36[cid36]["content"]
            identity36 = {
                "contract_sha256": score_hash36,
                "query_id": qid,
                "chunk_id": cid36,
                "question_sha256": question_hash36,
                "passage_sha256": hashlib.sha256(
                    passage36.encode("utf-8")
                ).hexdigest(),
            }
            cache_path36 = cache_folder36 / (
                digest(canonical(identity36)) + ".json"
            )

            valid36 = False
            if cache_path36.is_file():
                record36 = decode(cache_path36.read_bytes())
                body36 = record36["body"]

                require(
                    body36["identity"] == identity36
                    and record36["body_sha256"]
                    == digest(canonical(body36)),
                    "INVALID_CACHE_IDENTITY_OR_CHECKSUM:"
                    + str(cache_path36),
                )

                value36 = body36["score"]
                require(
                    type(value36) in (int, float)
                    and math.isfinite(value36),
                    "INVALID_CACHED_SCORE:" + str(cache_path36),
                )
                valid36 = True

            cached_flags36.append(valid36)

        for depth36 in (40, 100):
            candidate_set36 = set(order36[:depth36])
            available36 = len(candidate_set36 & gold36)
            ceiling36 = min(10, available36) / len(gold36)
            valid_count36 = sum(cached_flags36[:depth36])
            missing_count36 = depth36 - valid_count36

            counters36[depth36]["valid_cached_pairs"] += valid_count36
            counters36[depth36]["missing_pairs"] += missing_count36

            cases36.append({
                "query_id": qid,
                "source_types": golden36[qid]["source_types"],
                "depth": depth36,
                "baseline_recall_at_10": measured36["recall_at_10"],
                "recall_at_10_ceiling": ceiling36,
                "valid_cached_pairs": valid_count36,
                "missing_pairs": missing_count36,
            })

        if index36 % 100 == 0:
            print(json.dumps({
                "queries_checked": index36,
                "queries_total": 533,
                "new_model_pairs": 0,
            }), flush=True)

    before36 = c036_mean(baseline_metrics36.values())

    # The same 533-query baseline must reproduce C035 validation.
    recorded_base36 = report35_36["validation_metrics"]["baseline"]
    require(
        all(
            abs(before36[k] - recorded_base36[k]) <= 1e-9
            for k in before36
        ),
        "C035_VALIDATION_BASELINE_CONFLICT",
    )

    summaries36 = {}
    source_slices36 = {}

    for depth36 in (40, 100):
        rows36 = [r for r in cases36 if r["depth"] == depth36]
        require(len(rows36) == 533, "DEPTH_QUERY_COUNT_CONFLICT")

        ceiling36 = math.fsum(
            r["recall_at_10_ceiling"] for r in rows36
        ) / len(rows36)

        missing36 = counters36[depth36]["missing_pairs"]
        estimate_hours36 = missing36 * seconds_per_pair36 / 3600

        summaries36[str(depth36)] = {
            "queries": 533,
            "depth": depth36,
            "required_pairs": 533 * depth36,
            **counters36[depth36],
            "baseline_recall_at_10": before36["recall_at_10"],
            "recall_at_10_ceiling": ceiling36,
            "recall_target": 0.85,
            "ceiling_has_target_capacity": ceiling36 >= 0.85,
            "rough_hours_at_cached_runtime": estimate_hours36,
            "timing_reference_threads": cache_threads36,
            "runtime_estimate_scope": (
                "C027_WALL_TIME_PER_NEW_PAIR;"
                "DIFFERENT_QUESTIONS_AND_LENGTHS_MAY_CHANGE_SPEED"
            ),
        }

        source_slices36[str(depth36)] = {}
        for source36 in ("book", "transcript", "website"):
            source_rows36 = [
                r for r in rows36 if source36 in r["source_types"]
            ]
            if source_rows36:
                source_slices36[str(depth36)][source36] = {
                    "queries": len(source_rows36),
                    "recall_at_10_ceiling": math.fsum(
                        r["recall_at_10_ceiling"]
                        for r in source_rows36
                    ) / len(source_rows36),
                    "missing_pairs": sum(
                        r["missing_pairs"] for r in source_rows36
                    ),
                }

    if summaries36["40"]["ceiling_has_target_capacity"]:
        decision36 = (
            "TOP40_HAS_THEORETICAL_CAPACITY;"
            "REVIEW_COST_BEFORE_FIXED_METHOD_EVALUATION"
        )
    elif summaries36["100"]["ceiling_has_target_capacity"]:
        decision36 = (
            "TOP40_CANNOT_MEET_RECALL_TARGET;"
            "TOP100_HAS_CAPACITY_BUT_QUALITY_NOT_GUARANTEED"
        )
    else:
        decision36 = (
            "NEITHER_TOP40_NOR_TOP100_CAN_MEET_RECALL_TARGET;"
            "DO_NOT_START_BGE_FOR_TARGET_ACCEPTANCE_ON_THESE_INPUTS"
        )

    C036.update({
        "status": "COMPLETE",
        "query_count": 533,
        "baseline": before36,
        "depths": summaries36,
        "per_source": source_slices36,
        "score_contract_sha256": score_hash36,
        "timing_reference_report": C027["report"],
        "timing_seconds_per_new_pair": seconds_per_pair36,
        "timing_reference_threads": cache_threads36,
        "two_thread_runtime_measured": False,
        "runtime_contract_note": (
            "Cache counts apply to the existing score contract. "
            "Changing thread/runtime settings requires equivalence verification "
            "or a separate score contract; do not silently mix scores."
        ),
        "decision": decision36,
        "inference_started": False,
        "scope": "DEVELOPMENT_ORACLE_CAPACITY_AND_COST_ONLY",
        "elapsed_seconds": time.perf_counter() - started36,
    })

    C036["cases_artifact"] = save(
        "C036.bge_feasibility_cases", cases36, rows=True
    )
    C036["report"] = save(
        "C036.bge_feasibility_report", C036.copy()
    )

    html_rows36 = []
    for depth36 in ("40", "100"):
        result36 = summaries36[depth36]
        tick36 = (
            "✅" if result36["ceiling_has_target_capacity"] else ""
        )
        html_rows36.append(
            f"<tr><td>Top{depth36}</td>"
            f"<td>{result36['baseline_recall_at_10']:.2%}</td>"
            f"<td>{result36['recall_at_10_ceiling']:.2%}</td>"
            f"<td>85%</td><td>{tick36}</td>"
            f"<td>{result36['valid_cached_pairs']}</td>"
            f"<td>{result36['missing_pairs']}</td>"
            f"<td>{result36['rough_hours_at_cached_runtime']:.2f}</td>"
            "</tr>"
        )

    display(HTML(
        '<div dir="rtl" style="text-align:right">'
        "<p><strong>سقف نظری است؛ نتیجهٔ واقعی BGE نیست.</strong> "
        "تیک فقط ظرفیت نظری رسیدن به هدف را نشان می‌دهد.</p>"
        "<p>برآورد زمان مربوط به قرارداد قبلی "
        f"با {cache_threads36} رشته است؛ "
        "زمان دو رشته هنوز اندازه‌گیری نشده است.</p>"
        '<table style="text-align:center">'
        "<tr><th>ورودی</th><th>Recall فعلی baseline</th>"
        "<th>سقف Recall@10</th><th>هدف</th>"
        "<th>ظرفیت کافی</th><th>جفت معتبر در کش</th>"
        "<th>جفت جدید لازم</th><th>زمان تقریبی؛ ساعت</th></tr>"
        + "".join(html_rows36) + "</table></div>"
    ))

    print(json.dumps(C036, ensure_ascii=False, indent=2))

except Exception as exc:
    C036.update({"status": "FAILED", "error": str(exc)})
    print(json.dumps(C036, ensure_ascii=False, indent=2))
    raise
```

```python
# C037 v1 — Offline coverage and witness-rank diagnosis on Dev validation533.
# No neural model, network, new retrieval, candidate selection, or baseline change.

import json
import math
import collections
import statistics
from IPython.display import display, HTML

C037 = {
    "version": "NB22E_LARGE_POOL_COVERAGE_AND_RANK_TRACE_V1",
    "status": "PREPARING",
    "new_model_pairs": 0,
    "new_api_calls": 0,
    "downloads": False,
    "automatic_replacement": False,
    "release_acceptance_granted": False,
}


def c037_summary(rows):
    require(bool(rows), "EMPTY_SUMMARY")
    fields = (
        "top100_ceiling",
        "full_pool_ceiling",
        "word800_ceiling",
        "char800_ceiling",
        "word_char_union_ceiling",
        "hybrid100_ceiling",
        "pool_union_top100_ceiling",
        "pool_gold_missing_from_top100_fraction",
        "top100_gold_missing_from_pool_fraction",
    )
    result = {
        key: math.fsum(row[key] for row in rows) / len(rows)
        for key in fields
    }
    result.update({
        "queries": len(rows),
        "queries_with_pool_gold_outside_top100": sum(
            bool(row["pool_gold_outside_top100"]) for row in rows
        ),
        "queries_with_gold_outside_pool": sum(
            bool(row["gold_outside_pool"]) for row in rows
        ),
        "full_pool_has_recall_target_capacity": (
            result["full_pool_ceiling"] >= 0.85
        ),
    })
    return result


try:
    require(C036.get("status") == "COMPLETE", "C036_REQUIRED")

    report36_37 = read_json(C036["report"])
    plan35_37 = read_json(C035["plan_artifact"])
    ids37 = list(plan35_37["partitions"]["validation"])

    require(
        len(ids37) == len(set(ids37)) == 533,
        "EXPECTED_533_VALIDATION_QUERIES",
    )
    needed37 = set(ids37)

    print("بررسی هش فایل‌های pool و مسیرهای بازیابی…", flush=True)

    d_final37 = read_json(PINS["D_final"])
    verified37 = read_json(PINS["C056"])
    bindings37 = verified37["input_file_receipts"]

    # Confirm source corpus identity, without loading any vectors or model.
    chunks37 = keyed(read_rows(bindings37["chunks"]), "chunk_id")
    require(len(chunks37) == 3488, "CHUNK_POPULATION_CONFLICT")

    ranks37 = c025_subset(d_final37["retained_rankings"], needed37)
    pools37 = c025_subset(d_final37["pool_artifact"], needed37)
    hybrid37 = c025_subset(bindings37["hybrid"], needed37)

    channel_bindings37 = {}
    for binding in d_final37["artifacts"]:
        if binding["path"].split("/")[-1].startswith("A03.lexical_channels."):
            channel_bindings37[
                (binding["path"], binding_hash(binding))
            ] = binding

    require(
        len(channel_bindings37) == 1,
        "UNAMBIGUOUS_LEXICAL_ARTIFACT_REQUIRED",
    )
    channel_binding37 = next(iter(channel_bindings37.values()))
    channels37 = c025_subset(channel_binding37, needed37)

    cases37 = []
    rank_observations37 = {
        name: [] for name in ("baseline", "pool", "word", "char", "hybrid")
    }
    rank_missing37 = collections.Counter()
    witness_units37 = 0

    for index37, qid in enumerate(ids37, start=1):
        require(
            ranks37[qid]["split"]
            == pools37[qid]["split"]
            == hybrid37[qid]["split"]
            == "development",
            "SPLIT_CONFLICT",
        )

        baseline = ranks37[qid]["ranked_chunk_ids"]
        pool = pools37[qid]["candidate_chunk_ids"]
        channels = channels37[qid]["channels"]
        require(set(channels) == {"word", "char"}, "CHANNEL_NAMES_CONFLICT")

        hybrid_candidates = hybrid37[qid]["candidates"]
        require(
            len(hybrid_candidates) == 100
            and [r["rank"] for r in hybrid_candidates] == list(range(1, 101)),
            "HYBRID_RANK_SEQUENCE_CONFLICT",
        )
        hybrid = [row["chunk_id"] for row in hybrid_candidates]

        orders = {
            "baseline": baseline,
            "pool": pool,
            "word": channels["word"],
            "char": channels["char"],
            "hybrid": hybrid,
        }
        for name, order in orders.items():
            require(
                len(order) == len(set(order))
                and set(order).issubset(chunks37),
                "ORDER_IDENTITY_CONFLICT:" + name,
            )

        require(
            len(baseline) >= 100
            and 0 < len(pool) <= 800
            and len(channels["word"]) <= 800
            and len(channels["char"]) <= 800,
            "ORDER_DEPTH_CONFLICT",
        )

        gold = set(LAB["dev"]["gold"][qid])
        require(
            bool(gold) and gold.issubset(chunks37),
            "GOLD_BINDING_CONFLICT",
        )

        positions = {
            name: {cid: i + 1 for i, cid in enumerate(order)}
            for name, order in orders.items()
        }

        sets = {name: set(order) for name, order in orders.items()}
        top100 = set(baseline[:100])

        def ceiling(candidates):
            return min(10, len(candidates & gold)) / len(gold)

        trace = []
        for cid in sorted(gold):
            witness_units37 += 1
            ranks = {
                name: positions[name].get(cid) for name in orders
            }
            for name, rank in ranks.items():
                if rank is None:
                    rank_missing37[name] += 1
                else:
                    rank_observations37[name].append(rank)

            trace.append({
                "chunk_id": cid,
                "source_id": chunks37[cid].get("source_id"),
                "source_type": chunks37[cid]["source_type"],
                "ranks": ranks,
            })

        cases37.append({
            "query_id": qid,
            "source_types": sorted({
                chunks37[cid]["source_type"] for cid in gold
            }),
            "gold_count": len(gold),
            "pool_size": len(pool),
            "top100_ceiling": ceiling(top100),
            "full_pool_ceiling": ceiling(sets["pool"]),
            "word800_ceiling": ceiling(sets["word"]),
            "char800_ceiling": ceiling(sets["char"]),
            "word_char_union_ceiling": ceiling(
                sets["word"] | sets["char"]
            ),
            "hybrid100_ceiling": ceiling(sets["hybrid"]),
            "pool_union_top100_ceiling": ceiling(
                sets["pool"] | top100
            ),
            "pool_gold_missing_from_top100_fraction": (
                len((gold & sets["pool"]) - top100) / len(gold)
            ),
            "top100_gold_missing_from_pool_fraction": (
                len((gold & top100) - sets["pool"]) / len(gold)
            ),
            "pool_gold_outside_top100": sorted(
                (gold & sets["pool"]) - top100
            ),
            "gold_outside_pool": sorted(gold - sets["pool"]),
            "witness_rank_trace": trace,
        })

        if index37 % 100 == 0:
            print(json.dumps({
                "queries_checked": index37,
                "queries_total": 533,
                "new_model_pairs": 0,
            }), flush=True)

    overall37 = c037_summary(cases37)

    require(
        abs(
            overall37["top100_ceiling"]
            - report36_37["depths"]["100"]["recall_at_10_ceiling"]
        ) <= 1e-12,
        "C036_TOP100_CEILING_REPLAY_CONFLICT",
    )

    sources37 = {}
    for source in ("book", "transcript", "website"):
        rows = [r for r in cases37 if source in r["source_types"]]
        if rows:
            sources37[source] = c037_summary(rows)

    # Witness-unit distribution, not a query-macro performance metric.
    rank_distribution37 = {}
    for name in rank_observations37:
        ranks = rank_observations37[name]
        rank_distribution37[name] = {
            "witness_units_total": witness_units37,
            "present_units": len(ranks),
            "absent_units": rank_missing37[name],
            "median_rank_when_present": (
                statistics.median(ranks) if ranks else None
            ),
            "units_at_or_above_depth": {
                str(depth): sum(rank <= depth for rank in ranks)
                for depth in (10, 40, 100, 200, 400, 800)
            },
        }

    if overall37["full_pool_has_recall_target_capacity"]:
        next_step37 = (
            "FULL_POOL_HAS_ORACLE_CAPACITY;"
            "DESIGN_A_LABEL_BLIND_BOUNDED_SHORTLIST_FROM_EXISTING_CHANNELS;"
            "DO_NOT_SCORE_FULL_POOL_AUTOMATICALLY"
        )
    elif overall37["word_char_union_ceiling"] >= 0.85:
        next_step37 = (
            "SAVED_POOL_INSUFFICIENT_BUT_CHANNEL_UNION_HAS_CAPACITY;"
            "REVIEW_POOL_CONSTRUCTION_BEFORE_RERANKING"
        )
    else:
        next_step37 = (
            "SAVED_POOL_AND_WORD_CHAR_UNION_BELOW_TARGET;"
            "CANDIDATE_SELECTION_ALONE_CANNOT_FIX_THESE_INPUTS"
        )

    C037.update({
        "status": "COMPLETE",
        "query_count": 533,
        "overall": overall37,
        "per_source": sources37,
        "witness_rank_distribution": rank_distribution37,
        "source_bindings": {
            "baseline": d_final37["retained_rankings"],
            "pool": d_final37["pool_artifact"],
            "lexical_channels": channel_binding37,
            "hybrid": bindings37["hybrid"],
            "chunks": bindings37["chunks"],
        },
        "dense_rank_reconstructed": False,
        "gold_used_for_candidate_selection": False,
        "candidate_selection_produced": False,
        "next_step": next_step37,
        "scope": "DEVELOPMENT_DIAGNOSIS_ORACLE_CAPACITY_NOT_ACCEPTANCE",
    })

    C037["cases_artifact"] = save(
        "C037.pool_witness_rank_cases", cases37, rows=True
    )
    C037["report"] = save(
        "C037.pool_witness_rank_report", C037.copy()
    )

    html_rows37 = []
    groups37 = [("کل ۵۳۳ سؤال", overall37)]
    for name, label in (
        ("book", "کتاب"),
        ("transcript", "متن صوتی"),
        ("website", "وب‌سایت"),
    ):
        if name in sources37:
            groups37.append((label, sources37[name]))

    for label, result in groups37:
        tick = (
            "✅" if result["full_pool_has_recall_target_capacity"] else ""
        )
        html_rows37.append(
            f"<tr><td>{label}</td><td>{result['queries']}</td>"
            f"<td>{result['top100_ceiling']:.2%}</td>"
            f"<td>{result['full_pool_ceiling']:.2%}</td>"
            f"<td>{result['word800_ceiling']:.2%}</td>"
            f"<td>{result['char800_ceiling']:.2%}</td>"
            f"<td>{result['word_char_union_ceiling']:.2%}</td>"
            f"<td>{result['hybrid100_ceiling']:.2%}</td>"
            f"<td>85%</td><td>{tick}</td></tr>"
        )

    display(HTML(
        '<div dir="rtl" style="text-align:right">'
        "<p><strong>همهٔ درصدها سقف نظری Recall@10 هستند؛ "
        "نتیجهٔ واقعی رتبه‌بندی نیستند.</strong> "
        "تیک فقط ظرفیت pool را نشان می‌دهد. "
        "هدف ۸۵٪ الزام مستقل هر منبع نیست.</p>"
        '<table style="text-align:center">'
        "<tr><th>جمعیت</th><th>سؤال</th><th>Top100 baseline</th>"
        "<th>pool کامل</th><th>word تا ۸۰۰</th>"
        "<th>char تا ۸۰۰</th><th>اجتماع word/char</th>"
        "<th>Hybrid100</th><th>هدف</th><th>ظرفیت pool کافی</th></tr>"
        + "".join(html_rows37) + "</table></div>"
    ))

    print(json.dumps(C037, ensure_ascii=False, indent=2))

except Exception as exc:
    C037.update({"status": "FAILED", "error": str(exc)})
    print(json.dumps(C037, ensure_ascii=False, indent=2))
    raise
```

```python
# C038 v1 — Protected Top40 + existing F ridge shortlist.
# No retraining, BGE inference, downloads, or product replacement.

import json
import math
import time
import hashlib
import numpy as np
from IPython.display import display, HTML

C038 = {
    "version": "NB22E_LEARNED_PROTECTED_SHORTLIST_V1",
    "status": "PREPARING",
    "new_model_pairs": 0,
    "new_api_calls": 0,
    "downloads": False,
    "retraining": False,
    "automatic_replacement": False,
    "release_acceptance_granted": False,
}


def c038_select(baseline, pool, features, mu, sd, ridge):
    # No Gold labels or evaluation outcomes enter this function.
    standardized = (features - mu) / sd
    scores = ridge[0] + np.sum(
        standardized * ridge[1:], axis=1
    )
    require(np.isfinite(scores).all(), "NONFINITE_SHORTLIST_SCORES")

    selected = list(baseline[:40])
    seen = set(selected)
    indices = sorted(
        range(len(pool)),
        key=lambda i: (-float(scores[i]), pool[i]),
    )

    for i in indices:
        cid = pool[i]
        if cid not in seen:
            selected.append(cid)
            seen.add(cid)
        if len(selected) == 100:
            break

    return selected


def c038_summary(rows):
    require(bool(rows), "EMPTY_GROUP")
    result = {
        key: math.fsum(row[key] for row in rows) / len(rows)
        for key in (
            "top100_ceiling",
            "learned_shortlist_ceiling",
            "full_pool_ceiling",
            "delta_vs_top100",
        )
    }
    result.update({
        "queries": len(rows),
        "queries_better_than_top100": sum(
            r["delta_vs_top100"] > 1e-12 for r in rows
        ),
        "queries_worse_than_top100": sum(
            r["delta_vs_top100"] < -1e-12 for r in rows
        ),
        "queries_unchanged_vs_top100": sum(
            abs(r["delta_vs_top100"]) <= 1e-12 for r in rows
        ),
        "candidate_count_min": min(r["candidate_count"] for r in rows),
        "candidate_count_max": max(r["candidate_count"] for r in rows),
        "valid_cached_BGE_pairs": sum(r["valid_cached_BGE_pairs"] for r in rows),
        "missing_BGE_pairs": sum(r["missing_BGE_pairs"] for r in rows),
        "oracle_target_capacity": (
            result["learned_shortlist_ceiling"] >= 0.85
        ),
    })
    return result


try:
    started38 = time.perf_counter()
    require(C037.get("status") == "COMPLETE", "C037_REQUIRED")
    require(C033.get("status") == "COMPLETE", "C033_REQUIRED")

    report37_38 = read_json(C037["report"])
    cases37_38 = read_rows(C037["cases_artifact"])
    audit38 = read_json(C033["report"])
    f_report38 = read_json(audit38["F_report"])
    f_plan38 = read_json(f_report38["plan"])
    f_model38 = read_json(f_report38["models"])
    cost38 = read_json(C036["report"])

    reference38 = {}
    for row in cases37_38:
        qid = row["query_id"]
        require(qid not in reference38, "DUPLICATE_REFERENCE_QUERY")
        reference38[qid] = row

    ids38 = sorted(reference38)
    require(
        len(ids38) == 533
        and set(ids38) == set(f_plan38["partitions"]["validation"])
        and not set(ids38).intersection(f_plan38["partitions"]["fit"]),
        "FIT_VALIDATION_SEPARATION_CONFLICT",
    )

    require(
        binding_hash(f_plan38["baseline"])
        == binding_hash(report37_38["source_bindings"]["baseline"])
        and binding_hash(f_plan38["pool"])
        == binding_hash(report37_38["source_bindings"]["pool"]),
        "F_E_SOURCE_BINDING_CONFLICT",
    )

    print("بررسی مدل موجود، baseline و کش ویژگی‌ها…", flush=True)

    ranks38 = c025_subset(f_plan38["baseline"], set(ids38))
    pools38 = c025_subset(f_plan38["pool"], set(ids38))

    verified38 = read_json(PINS["C056"])
    chunks38 = keyed(
        read_rows(verified38["input_file_receipts"]["chunks"]), "chunk_id"
    )
    golden38 = keyed(
        read_rows(verified38["input_file_receipts"]["golden"]), "candidate_id"
    )

    expected_names38 = [
        "dense_cosine", "baseline_rr", "word_rr", "char_rr",
        "hybrid_rr", "word_coverage", "numeric_coverage",
        "term_frequency", "log_length",
        "baseline_present", "hybrid_present",
    ]
    require(
        f_model38["features"] == expected_names38,
        "F_FEATURE_SCHEMA_CONFLICT",
    )

    mu38 = np.asarray(f_model38["mu"], dtype=np.float64)
    sd38 = np.asarray(f_model38["sd"], dtype=np.float64)
    ridge38 = np.asarray(f_model38["ridge"], dtype=np.float64)

    require(
        mu38.shape == sd38.shape == (11,)
        and ridge38.shape == (12,)
        and np.all(sd38 > 0)
        and all(np.isfinite(a).all() for a in (mu38, sd38, ridge38)),
        "MODEL_PARAMETERS_CONFLICT",
    )

    feature_id38 = f_model38["feature_id"]
    feature_dir38 = path_for(
        "artifacts/phase2/quality_improvement/NB22F.v1/"
        "features_" + feature_id38
    )
    require(feature_dir38.is_dir(), "EXISTING_F_FEATURE_CACHE_REQUIRED")

    frozen38 = read_json(C020["method_artifact"])
    score_hash38 = frozen38["score_contract_sha256"]
    require(
        digest(canonical(frozen38["score_contract"])) == score_hash38
        and score_hash38 == cost38["score_contract_sha256"],
        "BGE_SCORE_CONTRACT_CONFLICT",
    )
    bge_cache38 = path_for(
        "artifacts/phase2/quality_improvement/"
        "NB22E.bge_top20_cache_v1/" + score_hash38
    )

    specification38 = {
        "method": "BASELINE_TOP40_PLUS_EXISTING_F_RIDGE60",
        "maximum_candidates": 100,
        "protected_prefix": 40,
        "selection_model": f_report38["models"],
        "model_feature_id": feature_id38,
        "tie_break": "CHUNK_ID",
        "selection_uses_gold": False,
        "retraining": False,
        "parameter_search": False,
        "evaluation_population": "F_VALIDATION_533",
    }
    C038["specification_artifact"] = save(
        "C038.learned_shortlist_spec", specification38
    )

    candidates38 = {}
    cases38 = []

    for index38, qid in enumerate(ids38, start=1):
        require(
            ranks38[qid]["split"] == pools38[qid]["split"] == "development",
            "DEVELOPMENT_SPLIT_REQUIRED",
        )
        baseline = ranks38[qid]["ranked_chunk_ids"]
        saved_pool = pools38[qid]["candidate_chunk_ids"]

        require(
            len(baseline) >= 100
            and len(baseline) == len(set(baseline))
            and len(saved_pool) == len(set(saved_pool))
            and set(baseline).issubset(chunks38)
            and set(saved_pool).issubset(chunks38),
            "INPUT_ORDER_CONFLICT",
        )

        # Reproduce NB22F's exact candidate order.
        model_pool = list(dict.fromkeys(baseline[:40] + saved_pool))
        feature_path = feature_dir38 / (
            hashlib.sha256(qid.encode()).hexdigest() + ".json"
        )
        require(
            feature_path.is_file(),
            "FEATURE_CACHE_MISSING_NO_RECOMPUTATION:" + qid,
        )
        record = decode(feature_path.read_bytes())
        payload = record["payload"]

        require(
            record["query_id"] == qid
            and record["feature_id"] == feature_id38
            and record["checksum"] == digest(canonical(payload))
            and payload["ids"] == model_pool,
            "FEATURE_CACHE_IDENTITY_CONFLICT:" + qid,
        )
        features = np.asarray(payload["x"], dtype=np.float64)
        require(
            features.shape == (len(model_pool), 11)
            and np.isfinite(features).all(),
            "FEATURE_ARRAY_CONFLICT",
        )

        selected = c038_select(
            baseline, model_pool, features, mu38, sd38, ridge38
        )
        require(
            selected[:40] == baseline[:40]
            and len(selected) == len(set(selected))
            and len(selected) <= 100
            and set(selected).issubset(model_pool),
            "SHORTLIST_CONTRACT_CONFLICT",
        )
        candidates38[qid] = selected

        # Labels are consulted only after selection.
        gold = set(LAB["dev"]["gold"][qid])
        require(bool(gold), "EMPTY_GOLD")

        def ceiling(candidates):
            return min(10, len(set(candidates) & gold)) / len(gold)

        top100_ceiling = ceiling(baseline[:100])
        pool_ceiling = ceiling(saved_pool)
        learned_ceiling = ceiling(selected)

        require(
            abs(top100_ceiling - reference38[qid]["top100_ceiling"]) <= 1e-12
            and abs(pool_ceiling - reference38[qid]["full_pool_ceiling"])
            <= 1e-12,
            "C037_CEILING_REPLAY_CONFLICT",
        )

        # Count valid existing scores; never fill a miss with inference.
        cached_count = 0
        question_hash = hashlib.sha256(
            golden38[qid]["evaluation_content"]["question"].encode("utf-8")
        ).hexdigest()

        for cid in selected:
            identity = {
                "contract_sha256": score_hash38,
                "query_id": qid,
                "chunk_id": cid,
                "question_sha256": question_hash,
                "passage_sha256": hashlib.sha256(
                    chunks38[cid]["content"].encode("utf-8")
                ).hexdigest(),
            }
            path = bge_cache38 / (digest(canonical(identity)) + ".json")
            if not path.is_file():
                continue

            cached = decode(path.read_bytes())
            body = cached["body"]
            require(
                body["identity"] == identity
                and cached["body_sha256"] == digest(canonical(body))
                and type(body["score"]) in (int, float)
                and math.isfinite(body["score"]),
                "INVALID_BGE_CACHE_RECORD",
            )
            cached_count += 1

        cases38.append({
            "query_id": qid,
            "source_types": reference38[qid]["source_types"],
            "candidate_count": len(selected),
            "top100_ceiling": top100_ceiling,
            "learned_shortlist_ceiling": learned_ceiling,
            "full_pool_ceiling": pool_ceiling,
            "delta_vs_top100": learned_ceiling - top100_ceiling,
            "valid_cached_BGE_pairs": cached_count,
            "missing_BGE_pairs": len(selected) - cached_count,
        })

        if index38 % 100 == 0:
            print(json.dumps({
                "queries_checked": index38,
                "queries_total": 533,
                "new_model_pairs": 0,
            }), flush=True)
        time.sleep(0.005)

    overall38 = c038_summary(cases38)
    source_slices38 = {}
    for source in ("book", "transcript", "website"):
        rows = [r for r in cases38 if source in r["source_types"]]
        if rows:
            source_slices38[source] = c038_summary(rows)

    estimated_hours38 = (
        overall38["missing_BGE_pairs"]
        * cost38["timing_seconds_per_new_pair"] / 3600
    )

    C038.update({
        "status": "COMPLETE",
        "overall": overall38,
        "per_source": source_slices38,
        "rough_BGE_hours_at_previous_runtime": estimated_hours38,
        "timing_reference_threads": cost38["timing_reference_threads"],
        "capacity_screen_passed": (
            overall38["oracle_target_capacity"]
            and overall38["delta_vs_top100"] > 1e-12
        ),
        "inference_started": False,
        "scope": "DEVELOPMENT_ORACLE_CAPACITY_NOT_QUALITY_ACCEPTANCE",
        "elapsed_seconds": time.perf_counter() - started38,
    })
    C038["candidate_artifact"] = save(
        "C038.learned_shortlist_candidates",
        {
            "specification": specification38,
            "candidates": candidates38,
        },
    )
    C038["cases_artifact"] = save(
        "C038.learned_shortlist_cases", cases38, rows=True
    )
    C038["report"] = save("C038.learned_shortlist_report", C038.copy())

    html_rows38 = []
    groups = [("کل جمعیت", overall38)]
    for source, label in (
        ("book", "کتاب"),
        ("transcript", "متن صوتی"),
        ("website", "وب‌سایت"),
    ):
        if source in source_slices38:
            groups.append((label, source_slices38[source]))

    for label, result in groups:
        tick = "✅" if result["oracle_target_capacity"] else ""
        html_rows38.append(
            f"<tr><td>{label}</td><td>{result['queries']}</td>"
            f"<td>{result['top100_ceiling']:.2%}</td>"
            f"<td>{result['learned_shortlist_ceiling']:.2%}</td>"
            f"<td>{result['full_pool_ceiling']:.2%}</td>"
            f"<td>{100*result['delta_vs_top100']:+.2f}</td>"
            f"<td>{result['queries_better_than_top100']}</td>"
            f"<td>{result['queries_worse_than_top100']}</td>"
            f"<td>85%</td><td>{tick}</td></tr>"
        )

    display(HTML(
        '<div dir="rtl" style="text-align:right">'
        "<p><strong>درصدها سقف نظری Recall@10 هستند؛ "
        "BGE اجرا نشده است.</strong> "
        "تیک ظرفیت نظری را نشان می‌دهد، نه قبولی واقعی. "
        "هدف منابع الزام پذیرش جداگانه نیست.</p>"
        '<table style="text-align:center">'
        "<tr><th>جمعیت</th><th>سؤال</th><th>Top100 فعلی</th>"
        "<th>انتخاب یادگرفته‌شده</th><th>pool کامل</th>"
        "<th>تغییر؛ واحد درصد</th><th>سؤال بهتر</th>"
        "<th>سؤال بدتر</th><th>هدف</th><th>ظرفیت کافی</th></tr>"
        + "".join(html_rows38) + "</table></div>"
    ))
    print(json.dumps(C038, ensure_ascii=False, indent=2))

except Exception as exc:
    C038.update({"status": "FAILED", "error": str(exc)})
    print(json.dumps(C038, ensure_ascii=False, indent=2))
    raise
```

<h3>C039 — نسخهٔ ۱: آموزش انتخاب نامزد با هدف عبور شاهد از مرز shortlist</h3>
<ul>
<li>حفظ Top40 و انتخاب حداکثر ۶۰ نامزد اضافه از pool موجود.</li>
<li>شروع از مدل ridge قبلی؛ آموزش فقط روی fit قبلی NB22F.</li>
<li>تمرکز آموزش روی شاهدهای خارج از Top40 و رقیبان مرز ورود به shortlist.</li>
<li>تنظیمات ثابت: هشت دور آموزش؛ بدون انتخاب دور یا وزن از validation.</li>
<li>ارزیابی پوشش روی همان ۵۳۳ سؤال validation؛ check اجرا نمی‌شود.</li>

<li>بدون BGE، دانلود، API، تغییر دادهٔ مبنا یا اعلام پذیرش ریلیز.</li>

```python
# C039 v1 — Boundary-aware shortlist training.
# One fixed experiment: no parameter sweep, BGE, or check evaluation.

import json
import math
import time
import hashlib
import numpy as np
from IPython.display import display, HTML
from threadpoolctl import threadpool_limits

C039 = {
    "version": "NB22E_BOUNDARY_AWARE_SHORTLIST_V1",
    "status": "PREPARING",
    "new_model_pairs": 0,
    "new_api_calls": 0,
    "downloads": False,
    "automatic_replacement": False,
    "release_acceptance_granted": False,
}

try:
    started39 = time.perf_counter()
    require(C038.get("status") == "COMPLETE", "C038_REQUIRED")

    audit39 = read_json(C033["report"])
    f_report39 = read_json(audit39["F_report"])
    f_plan39 = read_json(f_report39["plan"])
    f_model39 = read_json(f_report39["models"])

    fit_ids39 = list(f_plan39["partitions"]["fit"])
    val_ids39 = list(f_plan39["partitions"]["validation"])
    require(
        len(fit_ids39) == len(set(fit_ids39)) == 1507
        and len(val_ids39) == len(set(val_ids39)) == 533
        and not set(fit_ids39).intersection(val_ids39),
        "FROZEN_PARTITION_CONFLICT",
    )

    feature_id39 = f_model39["feature_id"]
    feature_folder39 = path_for(
        "artifacts/phase2/quality_improvement/NB22F.v1/"
        "features_" + feature_id39
    )

    mu39 = np.asarray(f_model39["mu"], dtype=np.float64)
    sd39 = np.asarray(f_model39["sd"], dtype=np.float64)
    original_ridge39 = np.asarray(f_model39["ridge"], dtype=np.float64)

    require(
        mu39.shape == sd39.shape == (11,)
        and original_ridge39.shape == (12,)
        and np.all(sd39 > 0)
        and all(np.isfinite(a).all()
                for a in (mu39, sd39, original_ridge39)),
        "MODEL_PARAMETER_CONFLICT",
    )

    needed39 = set(fit_ids39 + val_ids39)
    print("بررسی ورودی‌ها و ساخت جمعیت آموزش shortlist…", flush=True)
    ranks39 = c025_subset(f_plan39["baseline"], needed39)
    pools39 = c025_subset(f_plan39["pool"], needed39)

    def c039_features(qid):
        baseline = ranks39[qid]["ranked_chunk_ids"]
        require(
            ranks39[qid]["split"] == pools39[qid]["split"] == "development",
            "SPLIT_CONFLICT",
        )
        pool = list(dict.fromkeys(
            baseline[:40] + pools39[qid]["candidate_chunk_ids"]
        ))
        path = feature_folder39 / (
            hashlib.sha256(qid.encode()).hexdigest() + ".json"
        )
        require(path.is_file(), "EXISTING_FEATURE_CACHE_REQUIRED:" + qid)

        record = decode(path.read_bytes())
        payload = record["payload"]
        require(
            record["query_id"] == qid
            and record["feature_id"] == feature_id39
            and record["checksum"] == digest(canonical(payload))
            and payload["ids"] == pool,
            "FEATURE_CACHE_CONFLICT:" + qid,
        )

        x = np.asarray(payload["x"], dtype=np.float64)
        require(
            x.shape == (len(pool), 11) and np.isfinite(x).all(),
            "FEATURE_ARRAY_CONFLICT",
        )
        return baseline, pool, (x - mu39) / sd39

    spec39 = {
        "method": "PROTECTED_TOP40_BOUNDARY_AWARE_LINEAR_SHORTLIST",
        "maximum_candidates": 100,
        "protected_prefix": 40,
        "additional_slots": 60,
        "epochs": 8,
        "learning_rate": 0.02,
        "l2_to_ridge_anchor": 0.01,
        "gradient_norm_cap": 10.0,
        "seed": 2239,
        "training_objective": (
            "SOFTPLUS_MARGIN_BETWEEN_FIT_GOLD_AND_NEGATIVE_CUTOFF;"
            "QUERY_WEIGHTED_BY_INVERSE_TOTAL_GOLD_COUNT"
        ),
        "training_uses_only_fit": True,
        "selection_uses_gold": False,
        "parameter_search": False,
        "base_model": f_report39["models"],
        "feature_id": feature_id39,
    }
    C039["plan_artifact"] = save("C039.boundary_shortlist_plan", {
        "specification": spec39,
        "fit_ids": fit_ids39,
        "validation_ids": val_ids39,
        "baseline": f_plan39["baseline"],
        "pool": f_plan39["pool"],
        "check_evaluation_planned_in_this_cell": False,
    })

    # Keep only training queries with recoverable gold outside protected Top40.
    training39 = []
    fit_without_recoverable_tail_gold39 = 0

    for index, qid in enumerate(fit_ids39, start=1):
        baseline, pool, x = c039_features(qid)
        protected = set(baseline[:40])
        tail_indices = [i for i, cid in enumerate(pool) if cid not in protected]
        tail_x = x[tail_indices]
        tail_ids = [pool[i] for i in tail_indices]
        gold = set(LAB["dev"]["gold"][qid])
        require(bool(gold), "EMPTY_FIT_GOLD")

        positive = np.asarray(
            [i for i, cid in enumerate(tail_ids) if cid in gold],
            dtype=np.int64,
        )
        negative = np.asarray(
            [i for i, cid in enumerate(tail_ids) if cid not in gold],
            dtype=np.int64,
        )

        if len(positive) and len(negative) and len(tail_ids) > 60:
            training39.append({
                "x": tail_x,
                "positive": positive,
                "negative": negative,
                "gold_count": len(gold),
            })
        else:
            fit_without_recoverable_tail_gold39 += 1

        if index % 300 == 0:
            print(json.dumps({
                "fit_queries_checked": index,
                "fit_total": 1507,
                "training_queries": len(training39),
            }), flush=True)

    require(bool(training39), "NO_BOUNDARY_TRAINING_QUERIES")

    # Intercept cancels when comparing scores; only coefficients are trained.
    anchor39 = original_ridge39[1:].copy()
    w39 = anchor39.copy()
    rng39 = np.random.default_rng(2239)

    print("آموزش هشت دور ثابت؛ بدون استفاده از validation…", flush=True)

    with threadpool_limits(limits=2):
        for epoch in range(8):
            for group_index in rng39.permutation(len(training39)):
                item = training39[int(group_index)]
                x = item["x"]
                positive = item["positive"]
                negative = item["negative"]

                scores = np.sum(x * w39, axis=1)

                # To retain P positives among 60 slots, compare them with
                # the (61-P)-th highest negative when that cutoff exists.
                cutoff_index = max(0, 60 - len(positive))
                cutoff_index = min(cutoff_index, len(negative) - 1)
                negative_order = sorted(
                    negative.tolist(),
                    key=lambda i: (-float(scores[i]), i),
                )
                cutoff = negative_order[cutoff_index]

                differences = x[positive] - x[cutoff]
                margins = np.clip(
                    np.sum(differences * w39, axis=1), -30, 30
                )
                gradient = -np.sum(
                    differences / (1 + np.exp(margins))[:, None],
                    axis=0,
                ) / item["gold_count"]

                gradient += 0.01 * (w39 - anchor39)
                norm = float(np.linalg.norm(gradient))
                if norm > 10:
                    gradient *= 10 / norm

                w39 -= (0.02 / math.sqrt(epoch + 1)) * gradient

            require(np.isfinite(w39).all(), "NONFINITE_TRAINED_MODEL")
            print(json.dumps({
                "epoch_completed": epoch + 1,
                "epochs": 8,
                "training_queries": len(training39),
            }), flush=True)
            time.sleep(0.02)

    C039["model_artifact"] = save("C039.boundary_shortlist_model", {
        "specification": spec39,
        "mu": mu39.tolist(),
        "sd": sd39.tolist(),
        "anchor": anchor39.tolist(),
        "weights": w39.tolist(),
        "training_queries": len(training39),
        "fit_queries_without_training_pair": (
            fit_without_recoverable_tail_gold39
        ),
    })

    def c039_select(baseline, pool, standardized_features, weights):
        # No Gold labels are accepted by this selection function.
        selected = list(baseline[:40])
        seen = set(selected)
        scores = np.sum(standardized_features * weights, axis=1)
        require(np.isfinite(scores).all(), "NONFINITE_SELECTION_SCORE")

        for i in sorted(
            range(len(pool)),
            key=lambda i: (-float(scores[i]), pool[i]),
        ):
            if pool[i] not in seen:
                selected.append(pool[i])
                seen.add(pool[i])
            if len(selected) == 100:
                break
        return selected

    reference38_39 = {
        row["query_id"]: row
        for row in read_rows(C038["cases_artifact"])
    }
    require(
        set(reference38_39) == set(val_ids39),
        "C038_VALIDATION_COVERAGE_CONFLICT",
    )

    candidates39 = {}
    cases39 = []

    print("ارزیابی پوشش روی validation؛ BGE اجرا نمی‌شود…", flush=True)

    for index, qid in enumerate(val_ids39, start=1):
        baseline, pool, x = c039_features(qid)
        selected = c039_select(baseline, pool, x, w39)

        require(
            selected[:40] == baseline[:40]
            and len(selected) == len(set(selected))
            and len(selected) <= 100
            and set(selected).issubset(pool),
            "SHORTLIST_SELECTION_CONFLICT",
        )
        candidates39[qid] = selected

        # Evaluation labels are used only after selection is complete.
        gold = set(LAB["dev"]["gold"][qid])
        require(bool(gold), "EMPTY_VALIDATION_GOLD")

        def ceiling(ids):
            return min(10, len(set(ids) & gold)) / len(gold)

        top100 = ceiling(baseline[:100])
        measured = ceiling(selected)
        reference = reference38_39[qid]

        require(
            abs(top100 - reference["top100_ceiling"]) <= 1e-12,
            "TOP100_REFERENCE_CONFLICT",
        )
        cases39.append({
            "query_id": qid,
            "source_types": reference["source_types"],
            "top100_ceiling": top100,
            "previous_ridge_shortlist_ceiling": (
                reference["learned_shortlist_ceiling"]
            ),
            "boundary_shortlist_ceiling": measured,
            "delta_vs_top100": measured - top100,
            "candidate_count": len(selected),
        })

        if index % 100 == 0:
            print(json.dumps({
                "validation_queries_checked": index,
                "total": 533,
                "new_BGE_pairs": 0,
            }), flush=True)
        time.sleep(0.005)

    def summarize39(rows):
        require(bool(rows), "EMPTY_SUMMARY")
        result = {
            key: math.fsum(row[key] for row in rows) / len(rows)
            for key in (
                "top100_ceiling",
                "previous_ridge_shortlist_ceiling",
                "boundary_shortlist_ceiling",
                "delta_vs_top100",
            )
        }
        result.update({
            "queries": len(rows),
            "queries_better_than_top100": sum(
                r["delta_vs_top100"] > 1e-12 for r in rows
            ),
            "queries_worse_than_top100": sum(
                r["delta_vs_top100"] < -1e-12 for r in rows
            ),
            "oracle_target_capacity": (
                result["boundary_shortlist_ceiling"] >= 0.85
            ),
        })
        return result

    overall39 = summarize39(cases39)
    source_slices39 = {}
    for source in ("book", "transcript", "website"):
        source_rows = [r for r in cases39 if source in r["source_types"]]
        if source_rows:
            source_slices39[source] = summarize39(source_rows)

    C039.update({
        "status": "COMPLETE",
        "overall": overall39,
        "per_source": source_slices39,
        "capacity_screen_passed": (
            overall39["oracle_target_capacity"]
            and overall39["delta_vs_top100"] > 1e-12
        ),
        "inference_started": False,
        "check_evaluated": False,
        "elapsed_seconds": time.perf_counter() - started39,
        "scope": "EXPOSED_DEVELOPMENT_ORACLE_CAPACITY_NOT_ACCEPTANCE",
        "objective_limit": (
            "Boundary-aware surrogate; not an exact optimizer "
            "or a guarantee of Recall@100."
        ),
    })

    C039["candidate_artifact"] = save("C039.boundary_candidates", {
        "model": C039["model_artifact"],
        "candidates": candidates39,
    })
    C039["cases_artifact"] = save(
        "C039.boundary_coverage_cases", cases39, rows=True
    )
    C039["report"] = save("C039.boundary_coverage_report", C039.copy())

    html_rows39 = []
    groups = [("کل جمعیت", overall39)]
    for source, label in (
        ("book", "کتاب"),
        ("transcript", "متن صوتی"),
        ("website", "وب‌سایت"),
    ):
        if source in source_slices39:
            groups.append((label, source_slices39[source]))

    for label, result in groups:
        tick = "✅" if result["oracle_target_capacity"] else ""
        html_rows39.append(
            f"<tr><td>{label}</td><td>{result['queries']}</td>"
            f"<td>{result['top100_ceiling']:.2%}</td>"
            f"<td>{result['previous_ridge_shortlist_ceiling']:.2%}</td>"
            f"<td>{result['boundary_shortlist_ceiling']:.2%}</td>"
            f"<td>{100*result['delta_vs_top100']:+.2f}</td>"
            f"<td>85%</td><td>{tick}</td></tr>"
        )

    display(HTML(
        '<div dir="rtl" style="text-align:right">'
        "<p><strong>تمام اعداد، سقف نظری Recall@10 هستند؛ "
        "BGE اجرا نشده است.</strong> "
        "تیک فقط ظرفیت نظری است و پذیرش کیفیت نیست. "
        "هدف منابع الزام مستقل پذیرش نیست.</p>"
        '<table style="text-align:center">'
        "<tr><th>جمعیت</th><th>سؤال</th><th>Top100</th>"
        "<th>shortlist قبلی</th><th>shortlist مرزی</th>"
        "<th>تغییر نسبت به Top100؛ واحد درصد</th>"
        "<th>هدف</th><th>ظرفیت کافی</th></tr>"
        + "".join(html_rows39) + "</table></div>"
    ))
    print(json.dumps(C039, ensure_ascii=False, indent=2))

except Exception as exc:
    C039.update({"status": "FAILED", "error": str(exc)})
    print(json.dumps(C039, ensure_ascii=False, indent=2))
    raise
```

<h3>C040 — نسخهٔ ۱: آزمون محدود توان BGE روی شاهدهای عمیق</h3>
<ul>
<li>شش سؤال دارای شاهد داخل pool ولی خارج Top100؛ دو سؤال از هر منبع.</li>
<li>حفظ Top100، افزودن شاهدهای ثبت‌شدهٔ داخل pool و نامزدهای پخش‌شده در pool.</li>
<li>حداکثر ۲۰۰ نامزد در هر سؤال؛ استفاده از امتیازهای کش قبلی.</li>
<li>رتبه‌بندی BGE خالص برای سنجش توان مدل، بدون ترکیب وزن‌ها.</li>
<li>انتخاب نامزدها در این تشخیص از Golden استفاده می‌کند؛
این روش قابل استقرار یا ارزیابی پذیرش نیست.</li>
<li>حدود ۲۰ تا ۴۰ دقیقه با چهار رشته؛ نوبت‌های ادامه‌پذیر ۱۲دقیقه‌ای.</li>
<li>بدون دانلود و API؛ اجرای محلی BGE دارد.</li>
</ul>

```python
# C040 v1 — Gold-conditioned diagnostic of BGE ranking ability.
# NOT a deployable shortlist and NOT an acceptance evaluation.
# Local BGE inference; maximum 6 queries x 200 pairs.
# Resume by rerunning this cell after PARTIAL.

import json
import math
import time
import hashlib
import numpy as np
import torch
from IPython.display import display, HTML

C040 = {
    "version": "NB22E_DEEP_WITNESS_BGE_DIAGNOSTIC_V1",
    "status": "PREPARING",
    "downloads": False,
    "new_api_calls": 0,
    "new_model_pairs": 0,
    "gold_conditioned_diagnostic": True,
    "automatic_replacement": False,
    "release_acceptance_granted": False,
}

try:
    require(C037.get("status") == "COMPLETE", "C037_REQUIRED")
    require(globals().get("BGE_LOCAL_READY", False), "C015_REQUIRED")

    source_report40 = read_json(C037["report"])
    source_cases40 = read_rows(C037["cases_artifact"])
    frozen40 = read_json(C020["method_artifact"])

    contract40 = frozen40["score_contract"]
    score_hash40 = frozen40["score_contract_sha256"]
    require(
        digest(canonical(contract40)) == score_hash40
        and contract40["threads"] == 4
        and contract40["batch_size"] == 1
        and contract40["max_length"] == 512,
        "EXISTING_FOUR_THREAD_SCORE_CONTRACT_REQUIRED",
    )

    runtime40 = globals().get("BGE_RUNTIME")
    require(
        isinstance(runtime40, dict)
        and runtime40.get("model_id") == contract40["model_id"]
        and runtime40.get("revision") == contract40["revision"]
        and runtime40.get("device") == "cpu"
        and runtime40.get("dtype") == "float32"
        and torch.__version__ == contract40["torch"],
        "MATCHING_BGE_RUNTIME_REQUIRED_NO_AUTOMATIC_LOADING",
    )

    verified40 = read_json(PINS["C056"])
    chunks40 = keyed(
        read_rows(verified40["input_file_receipts"]["chunks"]), "chunk_id"
    )
    golden40 = keyed(
        read_rows(verified40["input_file_receipts"]["golden"]), "candidate_id"
    )

    # Explicitly enriched diagnostic, selected deterministically.
    selected40 = []
    for source in ("book", "transcript", "website"):
        eligible = [
            row for row in source_cases40
            if set(row["source_types"]) == {source}
            and row["pool_gold_outside_top100"]
        ]
        eligible.sort(key=lambda row: (
            hashlib.sha256(
                ("C040_DIAGNOSTIC_V1|" + row["query_id"]).encode()
            ).hexdigest(),
            row["query_id"],
        ))
        require(len(eligible) >= 2, "NOT_ENOUGH_DIAGNOSTIC_CASES:" + source)
        selected40.extend(eligible[:2])

    ids40 = [row["query_id"] for row in selected40]
    require(len(set(ids40)) == 6, "DIAGNOSTIC_QUERY_CONFLICT")

    baselines40 = c025_subset(
        source_report40["source_bindings"]["baseline"], set(ids40)
    )
    pools40 = c025_subset(
        source_report40["source_bindings"]["pool"], set(ids40)
    )

    candidate_map40 = {}
    deep_map40 = {}

    for case in selected40:
        qid = case["query_id"]
        baseline = baselines40[qid]["ranked_chunk_ids"]
        pool = pools40[qid]["candidate_chunk_ids"]
        gold = set(LAB["dev"]["gold"][qid])

        deep = (gold & set(pool)) - set(baseline[:100])
        require(
            deep == set(case["pool_gold_outside_top100"]),
            "DEEP_WITNESS_REPLAY_CONFLICT",
        )

        # Gold inclusion is deliberate and disclosed: this is diagnosis only.
        candidates = list(dict.fromkeys(
            baseline[:100] + sorted(gold & set(pool))
        ))
        require(len(candidates) <= 200, "DIAGNOSTIC_GOLD_EXCEEDS_BUDGET")

        # Deterministic coverage of the rest of the pool.
        remaining = [cid for cid in pool if cid not in set(candidates)]
        slots = min(200 - len(candidates), len(remaining))

        if slots:
            indices = np.linspace(
                0, len(remaining) - 1, slots, dtype=int
            )
            candidates.extend(remaining[int(i)] for i in indices)

        require(
            len(candidates) == len(set(candidates))
            and len(candidates) <= 200
            and deep.issubset(candidates)
            and set(candidates).issubset(chunks40),
            "DIAGNOSTIC_CANDIDATE_CONFLICT",
        )
        candidate_map40[qid] = candidates
        deep_map40[qid] = sorted(deep)

    C040["plan_artifact"] = save("C040.deep_witness_plan", {
        "query_ids": ids40,
        "candidates": candidate_map40,
        "deep_witnesses": deep_map40,
        "maximum_pairs": sum(map(len, candidate_map40.values())),
        "score_contract_sha256": score_hash40,
        "selection_uses_gold": True,
        "purpose": "RANKING_CAPABILITY_DIAGNOSTIC_ONLY",
        "session_seconds": 720,
        "threads": 4,
        "release_acceptance_granted": False,
    })

    model40 = runtime40["model"]
    tokenizer40 = runtime40["tokenizer"]
    model40.eval()
    torch.set_num_threads(4)

    cache_folder40 = path_for(
        "artifacts/phase2/quality_improvement/"
        "NB22E.bge_top20_cache_v1/" + score_hash40
    )
    cache_folder40.mkdir(parents=True, exist_ok=True)

    started40 = time.monotonic()
    new40 = 0
    cached40 = 0
    results40 = []
    orders40_diag = {}
    partial40 = False
    last_progress40 = started40

    print(json.dumps({
        "stage": "LOCAL_BGE_DEEP_WITNESS_DIAGNOSTIC",
        "queries": 6,
        "maximum_pairs": sum(map(len, candidate_map40.values())),
        "threads": 4,
        "gold_conditioned": True,
        "acceptance_evaluation": False,
        "downloads": False,
        "api_calls": 0,
    }), flush=True)

    for query_index40, qid in enumerate(ids40, start=1):
        question = golden40[qid]["evaluation_content"]["question"]
        scores = {}

        for cid in candidate_map40[qid]:
            if time.monotonic() - started40 >= 3600:
                partial40 = True
                break

            identity = {
                "contract_sha256": score_hash40,
                "query_id": qid,
                "chunk_id": cid,
                "question_sha256": hashlib.sha256(
                    question.encode("utf-8")
                ).hexdigest(),
                "passage_sha256": hashlib.sha256(
                    chunks40[cid]["content"].encode("utf-8")
                ).hexdigest(),
            }
            path = cache_folder40 / (
                digest(canonical(identity)) + ".json"
            )

            if path.is_file():
                record = decode(path.read_bytes())
                body = record["body"]
                require(
                    body["identity"] == identity
                    and record["body_sha256"] == digest(canonical(body)),
                    "CACHE_CONFLICT",
                )
                score = float(body["score"])
                cached40 += 1
            else:
                encoded = tokenizer40(
                    question,
                    chunks40[cid]["content"],
                    return_tensors="pt",
                    truncation=True,
                    max_length=512,
                    padding=False,
                )
                with torch.inference_mode():
                    logits = model40(**encoded).logits

                require(logits.numel() == 1, "INVALID_BGE_OUTPUT")
                score = float(logits.item())
                require(math.isfinite(score), "NONFINITE_BGE_SCORE")

                body = {
                    "identity": identity,
                    "score": score,
                    "input_tokens": int(encoded["input_ids"].shape[1]),
                }
                c017_atomic_json(path, {
                    "body": body,
                    "body_sha256": digest(canonical(body)),
                })
                new40 += 1
                C040["new_model_pairs"] = new40

            require(math.isfinite(score), "NONFINITE_CACHED_SCORE")
            scores[cid] = score

            if time.monotonic() - last_progress40 >= 30:
                print(json.dumps({
                    "query": query_index40,
                    "queries_total": 6,
                    "pairs_ready_for_query": len(scores),
                    "required_for_query": len(candidate_map40[qid]),
                    "new_pairs_this_session": new40,
                    "cached_pairs_this_session": cached40,
                    "elapsed_minutes": round(
                        (time.monotonic() - started40) / 60, 2
                    ),
                }), flush=True)
                last_progress40 = time.monotonic()

        if partial40:
            break

        position = {
            cid: i for i, cid in enumerate(candidate_map40[qid])
        }
        order = sorted(
            candidate_map40[qid],
            key=lambda cid: (-scores[cid], position[cid]),
        )
        rank = {cid: i + 1 for i, cid in enumerate(order)}
        gold = set(LAB["dev"]["gold"][qid])

        results40.append({
            "query_id": qid,
            "source_types": golden40[qid]["source_types"],
            "candidate_count": len(order),
            "deep_witness_count": len(deep_map40[qid]),
            "deep_witnesses_in_top10": sum(
                rank[cid] <= 10 for cid in deep_map40[qid]
            ),
            "deep_witness_ranks": {
                cid: rank[cid] for cid in deep_map40[qid]
            },
            "before_diagnostic": c034_metric(
                baselines40[qid]["ranked_chunk_ids"], gold
            ),
            "after_gold_conditioned_diagnostic": c034_metric(order, gold),
        })
        orders40_diag[qid] = order

        print(json.dumps({
            "queries_completed": len(results40),
            "total": 6,
            "deep_witness_ranks": results40[-1]["deep_witness_ranks"],
        }), flush=True)

    C040.update({
        "status": "PARTIAL_RESUMABLE" if partial40 else "COMPLETE",
        "completed_queries": len(results40),
        "expected_queries": 6,
        "results": results40,
        "new_model_pairs": new40,
        "cached_pairs_this_session": cached40,
        "session_seconds": time.monotonic() - started40,
        "scope": "GOLD_CONDITIONED_RANKING_DIAGNOSTIC_NOT_ACCEPTANCE",
    })

    if not partial40:
        C040["deep_witnesses_total"] = sum(
            r["deep_witness_count"] for r in results40
        )
        C040["deep_witnesses_recovered_in_top10"] = sum(
            r["deep_witnesses_in_top10"] for r in results40
        )
        C040["orders_artifact"] = save(
            "C040.deep_witness_orders", orders40_diag
        )

    C040["report"] = save("C040.deep_witness_report", C040.copy())

    if partial40:
        print(
            "نوبت ۱۲دقیقه‌ای تمام شد؛ کش حفظ شده است. "
            "برای ادامه فقط همین سل را دوباره اجرا کن.",
            flush=True,
        )

    print(json.dumps(C040, ensure_ascii=False, indent=2))

except KeyboardInterrupt:
    C040.update({
        "status": "INTERRUPTED_RESUMABLE",
        "message": "کش حفظ شده؛ برای ادامه همین سل را دوباره اجرا کن.",
    })
    print(json.dumps(C040, ensure_ascii=False, indent=2))
    raise

except Exception as exc:
    C040.update({"status": "FAILED", "error": str(exc)})
    print(json.dumps(C040, ensure_ascii=False, indent=2))
    raise
```
