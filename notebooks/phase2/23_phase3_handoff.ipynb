```python
# NB23.C001 — Central imports and bootstrap
# Scope: standard-library imports only; no API, filesystem writes, or Git operations.

import os
import json
import hashlib

from pathlib import Path
from datetime import datetime, timezone
from collections import Counter
```

```python
# NB23.C002 — One source selector -> one version-neutral NB23_INPUT (read-only, no API/Git/writes).
# Only this selector changes when switching a supported source. Unknown schemas STOP safely.
import os
import re
import json
import hashlib
from pathlib import Path

NB23_SOURCE_REF = (
    "reports/phase2/release/NB22.C054.quality_disposition/"
    "C054.supplemented_NB23_handoff."
    "f41ed8f332785201ec29a7e9ca6c7f60c122730044419d5b833745fca17c059f.json"
)
NB23_INPUT = None  # Prevent a stale previous version from surviving a failed rerun.


def nb23_c002_preflight():
    def require(ok, code):
        if not ok:
            raise RuntimeError(code)

    def hash_bytes(data):
        return hashlib.sha256(data).hexdigest()

    def canonical(value):
        return json.dumps(value, ensure_ascii=False, sort_keys=True,
                          separators=(",", ":"), allow_nan=False).encode("utf-8")

    def distinct(pairs):
        out = {}
        for key, value in pairs:
            require(key not in out, "DUPLICATE_JSON_KEY:" + key)
            out[key] = value
        return out

    def safe_reference(value):
        return (isinstance(value, str) and value.strip() and
                not Path(value).is_absolute() and ".." not in Path(value).parts)

    def find_pointers(node, depth=0):
        # Normalize upstream reference records into stable field names.
        found = []
        path_fields = ("path", "relative_path", "artifact_path", "file_path")
        digest_fields = ("file_bytes_sha256", "raw_byte_sha256", "raw_file_sha256",
                         "byte_sha256", "file_sha256")

        if depth > 5:
            return found

        if isinstance(node, str) and depth == 0 and node.endswith(".json"):
            found.append({"path": node, "declared_raw_sha256": None})

        elif isinstance(node, dict):
            for field in path_fields:
                path = node.get(field)

                if isinstance(path, str) and path.strip():
                    valid_hashes = [
                        node[k].lower()
                        for k in digest_fields
                        if isinstance(node.get(k), str)
                        and re.fullmatch(r"[a-fA-F0-9]{64}", node[k])
                    ]

                    require(
                        len(set(valid_hashes)) <= 1,
                        "CONFLICTING_DECLARED_RAW_HASHES:" + field
                    )

                    found.append({
                        "path": path,
                        "declared_raw_sha256": (
                            valid_hashes[0] if valid_hashes else None
                        )
                    })

            for child in node.values():
                if isinstance(child, (dict, list)):
                    found.extend(find_pointers(child, depth + 1))

        elif isinstance(node, list):
            for child in node[:100]:
                if isinstance(child, (dict, list)):
                    found.extend(find_pointers(child, depth + 1))

        seen, result = set(), []

        for candidate in found:
            require(
                safe_reference(candidate["path"]),
                "UNSAFE_UPSTREAM_EVIDENCE_PATH"
            )

            identity = (
                candidate["path"],
                candidate["declared_raw_sha256"]
            )

            if identity not in seen:
                seen.add(identity)
                result.append(candidate)

        return result

    def adapt_c054(obj):
        # Verified mapping for the known C054 handoff schema.
        required = (
            "checkpoint_disposition",
            "checkpoint_manifest",
            "checkpoint_validation",
            "release_gate"
        )

        require(
            isinstance(obj.get("inputs"), dict),
            "C054_INPUTS_NOT_OBJECT"
        )
        require(
            isinstance(obj.get("parent_handoff"), dict),
            "C054_PARENT_NOT_OBJECT"
        )
        require(
            all(k in obj["inputs"] for k in required),
            "C054_REQUIRED_INPUT_MISSING"
        )

        flags = (
            "release_acceptance_granted",
            "accepted_release_input_established",
            "action28_accepted_release_input_condition_met",
            "phase2_completion_claimed",
            "formal_NB23_start_authorized_by_this_handoff"
        )

        for field in flags:
            require(
                type(obj.get(field)) is bool,
                "C054_FLAG_MISSING_OR_INVALID:" + field
            )

        # This known C054 source is a non-accepted candidate.
        # An accepted release needs its own verified gate mapping.
        require(
            not any(obj[field] for field in flags),
            "ACCEPTED_RELEASE_REQUIRES_EXPLICIT_GATE_ADAPTER"
        )

        evidence = {
            key: find_pointers(obj["inputs"][key])
            for key in required
        }

        evidence["parent_handoff"] = find_pointers(
            obj["parent_handoff"]
        )

        require(
            all(evidence.values()),
            "C054_EVIDENCE_POINTER_NOT_FOUND"
        )

        return {
            "release_id": None,
            "release_accepted": False,
            "acceptance_evidence_verified": False,
            "handoff_mode": "DRAFT_ONLY_UNTIL_GATE_PROVEN",
            "evidence_refs": evidence,
            "manifest_refs": evidence["checkpoint_manifest"],
            "gate_refs": evidence["release_gate"],
            "quality_gate_status": "NOT_INDEPENDENTLY_VERIFIED_IN_C002",
            "source_status": (
                obj.get("status")
                if isinstance(obj.get("status"), str)
                else None
            ),
        }

    # The only registration point for NEW source schemas.
    # An identical verified schema can use this same adapter,
    # regardless of whether the producer is A, B or C.
    adapters = {
        (
            "NB22_C054_HANDOFF_QUALITY_SUPPLEMENT_V1",
            "NB22_TO_NB23_EVIDENCE_STATUS_HANDOFF"
        ): (
            "NB23_C054_SOURCE_ADAPTER_V1",
            adapt_c054
        )
    }

    report = {
        "cell": "NB23.C002",
        "action": 1,
        "package": "1_OF_2_SOURCE_PREFLIGHT_AND_NORMALIZATION",
        "source_ref": NB23_SOURCE_REF,
        "new_api_requests": 0,
        "files_written": 0,
        "git_changes": 0,
        "release_acceptance_granted": False,
        "normalized_input_ready": False
    }

    if (
        not safe_reference(NB23_SOURCE_REF)
        or Path(NB23_SOURCE_REF).suffix.lower() != ".json"
    ):
        return {
            **report,
            "status": "INVALID_SOURCE_REF"
        }, None

    markers = (
        Path("notebooks/phase2/checklist/NB23_phase3_handoff_checklist_fa.md"),
        Path("data/Road_Maps/RAG_finance_phase2_operational_roadmap.md")
    )

    override = os.getenv("NB23_REPO_ROOT", "").strip()

    candidates = (
        [Path(override).expanduser()] if override else []
    ) + [
        Path.cwd(),
        *Path.cwd().parents
    ]

    roots = {
        p.resolve()
        for p in candidates
        if all((p / marker).is_file() for marker in markers)
    }

    if len(roots) != 1:
        return {
            **report,
            "status": (
                "REPO_ROOT_NOT_FOUND"
                if not roots
                else "REPO_ROOT_AMBIGUOUS"
            ),
            "root_candidates_matching": len(roots)
        }, None

    root = roots.pop()
    source_file = (root / NB23_SOURCE_REF).resolve()

    if (
        root not in source_file.parents
        or not source_file.is_file()
    ):
        return {
            **report,
            "status": "SOURCE_NOT_FOUND_OR_OUTSIDE_REPO"
        }, None

    if not 0 < source_file.stat().st_size <= 32 * 1024 * 1024:
        return {
            **report,
            "status": "SOURCE_SIZE_OUT_OF_SCOPE"
        }, None

    raw = source_file.read_bytes()
    source_sha = hash_bytes(raw)

    try:
        obj = json.loads(
            raw.decode("utf-8-sig"),
            object_pairs_hook=distinct,
            parse_constant=lambda x: (
                _ for _ in ()
            ).throw(ValueError(x))
        )

        require(
            isinstance(obj, dict),
            "SOURCE_JSON_NOT_OBJECT"
        )

        canonical_sha = hash_bytes(canonical(obj))

    except (
        ValueError,
        UnicodeDecodeError,
        TypeError,
        RuntimeError
    ) as exc:
        return {
            **report,
            "status": "SOURCE_JSON_INVALID",
            "error_code": str(exc)[:100]
        }, None

    schema_key = (
        obj.get("schema_version"),
        obj.get("artifact_type")
    )

    report.update({
        "source_available": True,
        "repo_folder_name": root.name,
        "source_bytes": len(raw),
        "source_raw_byte_sha256": source_sha,
        "source_json_canonical_sha256": canonical_sha,
        "observed_schema_version": schema_key[0],
        "observed_artifact_type": schema_key[1]
    })

    if (
        not all(isinstance(x, str) for x in schema_key)
        or schema_key not in adapters
    ):
        return {
            **report,
            "status": "ADAPTER_REQUIRED_UNRECOGNIZED_SOURCE_SCHEMA",
            "adapter_state": "UNSUPPORTED_NOT_GUESSED"
        }, None

    adapter_version, adapter = adapters[schema_key]

    name_hash = re.search(
        r"\.([a-f0-9]{64})\.json$",
        source_file.name
    )

    require(
        name_hash is None
        or name_hash.group(1) == source_sha,
        "SOURCE_FILENAME_BYTE_SHA256_MISMATCH"
    )

    mapped = adapter(obj)

    # Preserve the exact fingerprint algorithm used in C003-C012.
    identity = {
        "source_ref": NB23_SOURCE_REF,
        "source_raw_byte_sha256": source_sha,
        "source_schema_version": schema_key[0],
        "source_artifact_type": schema_key[1],
        "adapter_version": adapter_version
    }

    normalized = {
        "contract_version": "nb23.source.v1",
        "source_ref": NB23_SOURCE_REF,
        "source_raw_byte_sha256": source_sha,
        "source_json_canonical_sha256": canonical_sha,
        "source_schema_version": schema_key[0],
        "source_artifact_type": schema_key[1],
        "adapter_version": adapter_version,
        "source_identity": identity,
        "source_fingerprint": hash_bytes(
            canonical(identity)
        ),
        **mapped
    }

    # Every supported adapter must return these stable fields.
    required_keys = (
        "contract_version",
        "source_ref",
        "source_fingerprint",
        "source_identity",
        "adapter_version",
        "release_id",
        "release_accepted",
        "acceptance_evidence_verified",
        "handoff_mode",
        "manifest_refs",
        "gate_refs",
        "evidence_refs"
    )

    require(
        all(k in normalized for k in required_keys),
        "NORMALIZED_ADAPTER_MISSING_FIELD"
    )

    require(
        type(normalized["release_accepted"]) is bool
        and type(normalized["acceptance_evidence_verified"]) is bool,
        "NORMALIZED_RELEASE_FLAGS_INVALID"
    )

    report.update({
        "status": "SOURCE_NORMALIZED_DRAFT",
        "adapter_state": "SUPPORTED",
        "adapter_version": adapter_version,
        "source_fingerprint": normalized["source_fingerprint"],
        "normalized_contract_version": normalized["contract_version"],
        "normalized_input_ready": True,
        "release_accepted": normalized["release_accepted"],
        "acceptance_evidence_verified": normalized[
            "acceptance_evidence_verified"
        ],
        "normalized_evidence_groups": sorted(
            normalized["evidence_refs"]
        ),
        "next_step": (
            "C003 must consume NB23_INPUT, "
            "not parse source schema again"
        )
    })

    return report, normalized


nb23_c002_result, NB23_INPUT = nb23_c002_preflight()

print(json.dumps(
    nb23_c002_result,
    ensure_ascii=False,
    indent=2
))

if NB23_INPUT is None:
    raise RuntimeError(
        "NB23_C002_STOP:" + nb23_c002_result["status"]
    )
```

```python
# NB23.C012 — Action 10: version-neutral final readiness audit and immutable V2 evidence index.
# Uses NB23_INPUT and exact C003-C011 V2 receipts; independent synthetic regression, never grants acceptance.
import os
import re
import json
import hashlib
import tempfile
from pathlib import Path
from jsonschema import Draft202012Validator


def nb23_c012_final_readiness():
    def require(ok, reason):
        if not ok:
            raise RuntimeError(reason)

    def sha(raw):
        return hashlib.sha256(raw).hexdigest()

    def canon(value):
        return json.dumps(value, ensure_ascii=False, sort_keys=True,
                          separators=(",", ":"), allow_nan=False).encode("utf-8")

    def unique(pairs):
        obj = {}
        for key, value in pairs:
            require(key not in obj, "DUPLICATE_JSON_KEY:" + key)
            obj[key] = value
        return obj

    def decode(raw):
        return json.loads(raw.decode("utf-8-sig"), object_pairs_hook=unique,
                          parse_constant=lambda x: (_ for _ in ()).throw(ValueError(x)))

    def immutable(path, raw):
        path.parent.mkdir(parents=True, exist_ok=True)
        fd, temporary = tempfile.mkstemp(dir=path.parent, prefix=".NB23.C012N.")
        try:
            with os.fdopen(fd, "wb") as stream:
                stream.write(raw)
                stream.flush()
                os.fsync(stream.fileno())
            try:
                os.link(temporary, path)
            except FileExistsError:
                require(path.read_bytes() == raw, "IMMUTABLE_OUTPUT_COLLISION:" + path.name)
        finally:
            if os.path.exists(temporary):
                os.unlink(temporary)
        require(path.read_bytes() == raw, "IMMUTABLE_RELOAD_MISMATCH:" + path.name)

    require("NB23_INPUT" in globals() and isinstance(NB23_INPUT, dict),
            "C002_NORMALIZED_SOURCE_MISSING")
    require("NB23_SOURCE_REF" in globals() and NB23_SOURCE_REF == NB23_INPUT.get("source_ref"),
            "SOURCE_SWITCH_DETECTED_RERUN_FROM_C002")
    require(NB23_INPUT.get("contract_version") == "nb23.source.v1",
            "NORMALIZED_SOURCE_CONTRACT_UNSUPPORTED")

    for number in ("003", "004", "005", "006", "007", "008", "009", "010", "011"):
        name = "nb23_c" + number + "_result"
        require(name in globals() and isinstance(globals()[name], dict),
                "UPSTREAM_RESULT_MISSING:" + name)

    selected = NB23_INPUT["source_ref"]
    identity = NB23_INPUT.get("source_identity")

    require(isinstance(identity, dict) and identity == {
        "source_ref": selected,
        "source_raw_byte_sha256": NB23_INPUT["source_raw_byte_sha256"],
        "source_schema_version": NB23_INPUT["source_schema_version"],
        "source_artifact_type": NB23_INPUT["source_artifact_type"],
        "adapter_version": NB23_INPUT["adapter_version"],
    }, "SOURCE_IDENTITY_INVALID")

    fp = sha(canon(identity))

    require(
        fp == NB23_INPUT.get("source_fingerprint")
        and all(
            globals()["nb23_c" + n + "_result"].get("source_fingerprint") == fp
            for n in ("003", "004", "005", "006", "007", "008", "009", "010", "011")
        ),
        "UPSTREAM_SOURCE_FINGERPRINT_CONFLICT"
    )

    accepted = NB23_INPUT.get("release_accepted")
    acceptance_proof = NB23_INPUT.get("acceptance_evidence_verified")
    release_id = NB23_INPUT.get("release_id")

    require(
        type(accepted) is bool
        and type(acceptance_proof) is bool
        and (
            (not accepted and release_id is None)
            or (
                accepted
                and acceptance_proof
                and isinstance(release_id, str)
                and bool(release_id.strip())
            )
        ),
        "NORMALIZED_RELEASE_ACCEPTANCE_INCONSISTENT"
    )

    marker = Path("notebooks/phase2/checklist/NB23_phase3_handoff_checklist_fa.md")
    override = os.getenv("NB23_REPO_ROOT", "").strip()
    locations = (
        [Path(override).expanduser()] if override else []
    ) + [Path.cwd(), *Path.cwd().parents]

    roots = {
        p.resolve()
        for p in locations
        if (p / marker).is_file()
    }

    require(len(roots) == 1, "REPOSITORY_ROOT_MISSING_OR_AMBIGUOUS")
    root = roots.pop()
    work = (root / "artifacts/phase2/phase3_handoff/work" / fp).resolve()

    verified = {}

    def checked_file(relative, expected=None):
        require(
            isinstance(relative, str) and bool(relative.strip()),
            "EVIDENCE_REF_EMPTY"
        )

        ref = Path(relative)

        require(
            not ref.is_absolute() and ".." not in ref.parts,
            "UNSAFE_EVIDENCE_PATH:" + relative
        )

        path = (root / ref).resolve()

        require(
            root in path.parents and path.is_file(),
            "EVIDENCE_FILE_MISSING:" + relative
        )

        raw = path.read_bytes()
        actual = sha(raw)

        if expected is not None:
            require(
                isinstance(expected, str) and actual == expected,
                "EVIDENCE_SHA_MISMATCH:" + relative
            )

        normalized = path.relative_to(root).as_posix()

        if normalized in verified:
            require(
                verified[normalized]["file_bytes_sha256"] == actual,
                "EVIDENCE_CHANGED_DURING_AUDIT:" + normalized
            )
        else:
            verified[normalized] = {
                "path": normalized,
                "file_bytes_sha256": actual,
                "bytes": len(raw)
            }

        return raw

    selected_raw = checked_file(
        selected,
        NB23_INPUT["source_raw_byte_sha256"]
    )

    require(
        sha(selected_raw) == identity["source_raw_byte_sha256"],
        "SOURCE_BYTES_CHANGED_SINCE_C002"
    )

    source = decode(selected_raw)

    require(
        isinstance(source, dict)
        and source.get("schema_version") == identity["source_schema_version"]
        and source.get("artifact_type") == identity["source_artifact_type"],
        "SELECTED_SOURCE_IDENTITY_CHANGED"
    )

    # Exact receipt identities; never glob for an arbitrary old version.
    specs = {
        "C003": (
            "NB23.C003.normalized_source_binding",
            "artifact_path", "artifact_sha256"
        ),
        "C004": (
            "NB23.C004.normalized_request_contract_test",
            "report_path", "report_sha256"
        ),
        "C005": (
            "NB23.C005.normalized_response_contract_test",
            "report_path", "report_sha256"
        ),
        "C006": (
            "NB23.C006.normalized_citation_resolver_test",
            "report_path", "report_sha256"
        ),
        "C007": (
            "NB23.C007.normalized_status_error_contract_test",
            "report_path", "report_sha256"
        ),
        "C008_lock": (
            "NB23.C008.normalized_source_version_lock",
            "source_version_lock_path", "source_version_lock_sha256"
        ),
        "C008_report": (
            "NB23.C008.normalized_version_trace_test",
            "report_path", "report_sha256"
        ),
        "C009": (
            "NB23.C009.normalized_contract_test_report",
            "report_path", "report_sha256"
        ),
        "C010_matrix": (
            "NB23.C010.normalized_backward_compatibility_matrix",
            "matrix_path", "matrix_sha256"
        ),
        "C010_report": (
            "NB23.C010.normalized_compatibility_drift_test",
            "report_path", "report_sha256"
        ),
        "C011": (
            "NB23.C011.normalized_consumer_documentation_test",
            "report_path", "report_sha256"
        ),
    }

    receipts, hashes, paths = {}, {}, {}

    for key, (prefix, path_key, hash_key) in specs.items():
        number = (
            "008" if key.startswith("C008")
            else "010" if key.startswith("C010")
            else key[1:]
        )

        result = globals()["nb23_c" + number + "_result"]
        relative = result.get(path_key)
        expected = result.get(hash_key)

        require(
            isinstance(expected, str)
            and re.fullmatch(r"[0-9a-f]{64}", expected),
            "RECEIPT_SHA_MISSING:" + key
        )

        path = (
            (root / relative).resolve()
            if isinstance(relative, str) else root
        )

        require(
            path.parent == work
            and path.name == prefix + "." + expected + ".json",
            "RECEIPT_NAME_OR_VERSION_MISMATCH:" + key
        )

        raw = checked_file(relative, expected)
        obj = decode(raw)

        require(isinstance(obj, dict), "RECEIPT_NOT_OBJECT:" + key)

        receipts[key] = obj
        hashes[key] = expected
        paths[key] = relative

    c3, c4, c5, c6, c7 = [
        receipts[k]
        for k in ("C003", "C004", "C005", "C006", "C007")
    ]

    lock, c8, c9 = [
        receipts[k]
        for k in ("C008_lock", "C008_report", "C009")
    ]

    matrix = receipts["C010_matrix"]
    c10 = receipts["C010_report"]
    c11 = receipts["C011"]

    def expect_schema(item, expected, key):
        require(
            item.get("schema_version") == expected
            and item.get("source_fingerprint") == fp,
            "NORMALIZED_RECEIPT_SCHEMA_OR_SOURCE_CONFLICT:" + key
        )

    for key, schema_version in (
        ("C003", "NB23_C003_NORMALIZED_SOURCE_BINDING_RECEIPT_V2"),
        ("C004", "NB23_C004_NORMALIZED_REQUEST_CONTRACT_TEST_REPORT_V2"),
        ("C005", "NB23_C005_NORMALIZED_RESPONSE_CONTRACT_REPORT_V2"),
        ("C006", "NB23_C006_NORMALIZED_CITATION_RESOLVER_REPORT_V2"),
        ("C007", "NB23_C007_NORMALIZED_STATUS_ERROR_CONTRACT_REPORT_V2"),
        ("C008_lock", "NB23_C008_NORMALIZED_SOURCE_VERSION_LOCK_V2"),
        ("C008_report", "NB23_C008_NORMALIZED_VERSION_TRACE_TEST_REPORT_V2"),
        ("C009", "NB23_C009_NORMALIZED_CONTRACT_TEST_REPORT_V2"),
        ("C010_matrix", "NB23_C010_NORMALIZED_BACKWARD_COMPATIBILITY_MATRIX_V2"),
        ("C010_report", "NB23_C010_NORMALIZED_COMPATIBILITY_DRIFT_REPORT_V2"),
        ("C011", "NB23_C011_NORMALIZED_CONSUMER_DOCUMENTATION_REPORT_V2"),
    ):
        expect_schema(receipts[key], schema_version, key)

    require(
        c3.get("source_identity") == identity
        and c3.get("input_evidence_integrity") == "ALL_TYPED_BYTES_VERIFIED"
        and c3.get("unresolved_binding_groups") == []
        and c3.get("release_accepted") is accepted
        and c3.get("acceptance_evidence_verified") is acceptance_proof
        and c3.get("release_id") == release_id,
        "C003_EVIDENCE_OR_GATE_CONFLICT"
    )

    require(
        c4.get("C003_receipt_sha256") == hashes["C003"]
        and c4.get("passed_fixtures") == 12
        and c4.get("failed_fixtures") == 0
        and c4.get("upstream_release_accepted") is accepted,
        "C004_CHAIN_OR_TEST_CONFLICT"
    )

    require(
        c5.get("C003_receipt_file_sha256") == hashes["C003"]
        and c5.get("C004_receipt_file_sha256") == hashes["C004"]
        and c5.get("fixture_pass_count") == 20
        and c5.get("fixture_fail_count") == 0
        and c5.get("release_id_from_normalized_input") == release_id
        and c5.get("upstream_release_accepted") is accepted,
        "C005_CHAIN_OR_TEST_CONFLICT"
    )

    require(
        c6.get("input_receipt_sha256") == {
            k: hashes[k] for k in ("C003", "C004", "C005")
        }
        and c6.get("tests_passed") == 17
        and c6.get("tests_failed") == 0
        and c6.get("release_id_from_normalized_input") == release_id
        and c6.get("upstream_release_accepted") is accepted,
        "C006_CHAIN_OR_TEST_CONFLICT"
    )

    require(
        c7.get("parent_receipts_sha256") == {
            k: hashes[k] for k in ("C003", "C004", "C005", "C006")
        }
        and c7.get("fixtures_passed") == 38
        and c7.get("fixtures_failed") == 0
        and c7.get("release_id_from_normalized_input") == release_id
        and c7.get("upstream_release_accepted") is accepted,
        "C007_CHAIN_OR_TEST_CONFLICT"
    )

    require(
        lock.get("source_identity") == identity
        and lock.get("previous_contract_receipts_sha256") == {
            k: hashes[k]
            for k in ("C003", "C004", "C005", "C006", "C007")
        }
        and lock.get("release_id") == release_id
        and lock.get("release_is_accepted") is accepted
        and lock.get("acceptance_evidence_verified_by_source_adapter")
        is acceptance_proof,
        "C008_LOCK_CHAIN_OR_ACCEPTANCE_CONFLICT"
    )

    require(
        c8.get("source_version_lock_sha256") == hashes["C008_lock"]
        and c8.get("parent_receipts_sha256") == {
            k: hashes[k]
            for k in ("C003", "C004", "C005", "C006", "C007")
        }
        and c8.get("passed") == 12
        and c8.get("failed") == 0
        and c8.get("release_id_from_normalized_input") == release_id,
        "C008_REPORT_CHAIN_OR_TEST_CONFLICT"
    )

    require(
        c9.get("parent_receipt_hashes") == {
            k: hashes[k]
            for k in (
                "C003", "C004", "C005", "C006", "C007",
                "C008_lock", "C008_report"
            )
        }
        and c9.get("tests_passed") == 41
        and c9.get("tests_failed") == 0
        and c9.get("release_id_from_normalized_input") == release_id
        and c9.get("accepted_for_phase3") is False,
        "C009_FIXTURE_CHAIN_OR_TEST_CONFLICT"
    )

    require(
        matrix.get("compatibility_decision") ==
        "INITIAL_DRAFT_NO_VERIFIED_PREVIOUS_CONSUMER_SCHEMA"
        and matrix.get("backward_compatibility_pass_claimed") is False
        and matrix.get("release_id_from_normalized_input") == release_id,
        "C010_MATRIX_STATE_CONFLICT"
    )

    require(
        c10.get("compatibility_matrix_sha256") == hashes["C010_matrix"]
        and c10.get("parent_receipt_sha256") == {
            k: hashes[k]
            for k in (
                "C003", "C004", "C005", "C006", "C007",
                "C008_lock", "C008_report", "C009"
            )
        }
        and c10.get("hypothetical_tests_passed") == 19
        and c10.get("actual_backward_compatibility_verified") is False
        and c10.get("release_id_from_normalized_input") == release_id,
        "C010_REPORT_CHAIN_OR_TEST_CONFLICT"
    )

    require(
        c11.get("parent_receipt_sha256") == {
            k: hashes[k]
            for k in (
                "C003", "C004", "C005", "C006", "C007",
                "C008_lock", "C008_report", "C009",
                "C010_matrix", "C010_report"
            )
        }
        and c11.get("conversation_count") == 12
        and c11.get("turn_count") == 13
        and c11.get("tests_passed") == 23
        and c11.get("tests_failed") == 0
        and c11.get("release_id_from_normalized_input") == release_id
        and c11.get("all_samples_are_synthetic") is True
        and c11.get("accepted_for_phase3") is False,
        "C011_DOCUMENTATION_CHAIN_OR_TEST_CONFLICT"
    )

    # Independently verify all source-normalized evidence references.
    bindings = c3.get("input_bindings")

    required_groups = {
        "checkpoint_disposition",
        "checkpoint_manifest",
        "checkpoint_validation",
        "release_gate",
        "parent_handoff"
    }

    require(
        isinstance(bindings, list) and len(bindings) >= 5,
        "C003_VERIFIED_BINDINGS_MISSING"
    )

    grouped, role_paths = {}, {}

    for row in bindings:
        require(
            isinstance(row, dict)
            and row.get("status") == "BYTE_HASH_VERIFIED",
            "SOURCE_INPUT_NOT_BYTE_VERIFIED"
        )

        group = row["group"]
        path = row["path"]

        checked_file(path, row["actual_file_bytes_sha256"])

        grouped.setdefault(group, []).append({
            "path": path,
            "declared_raw_sha256": row["declared_raw_sha256"]
        })

        role_paths.setdefault(group, []).append(path)

    require(
        required_groups.issubset(grouped)
        and grouped == NB23_INPUT.get("evidence_refs")
        and grouped["checkpoint_manifest"] == NB23_INPUT.get("manifest_refs")
        and grouped["release_gate"] == NB23_INPUT.get("gate_refs"),
        "SOURCE_NORMALIZED_EVIDENCE_MISMATCH"
    )

    require(
        len(role_paths["release_gate"]) == 1
        and len(role_paths["checkpoint_manifest"]) == 1,
        "MULTIPLE_RELEASE_GATE_OR_MANIFEST_REQUIRES_ADAPTER"
    )

    upstream_gate = decode(
        checked_file(role_paths["release_gate"][0])
    )

    require(
        isinstance(upstream_gate, dict),
        "UPSTREAM_GATE_NOT_OBJECT"
    )

    gate_status = upstream_gate.get("status")

    require(
        gate_status is None or isinstance(gate_status, str),
        "UPSTREAM_GATE_STATUS_TYPE_UNEXPECTED"
    )

    require(
        c3.get("gate_refs") == NB23_INPUT["gate_refs"]
        and c3.get("manifest_refs") == NB23_INPUT["manifest_refs"]
        and lock.get("source_acceptance_gate") == gate_status,
        "LOCK_GATE_OR_MANIFEST_BINDING_CHANGED"
    )

    unresolved = lock.get("authoritative_components_unresolved")

    require(
        isinstance(unresolved, list)
        and len(unresolved) == len(set(unresolved))
        and set(unresolved).issubset({
            "code",
            "dataset_chunk",
            "embedding_index",
            "retrieval_reranker_context",
            "model_prompt",
            "policy",
            "evaluation"
        }),
        "AUTHORITATIVE_COMPONENT_GAP_LEDGER_INVALID"
    )

    # Independently reload NB07 lineage and the four common contracts.
    checked_file(
        c4["generator_reference"]["path"],
        c4["generator_reference"]["file_bytes_sha256"]
    )

    nb07_raw = checked_file(
        "reports/phase2/unified_schema_lineage_acceptance_v1.json",
        c6["nb07_acceptance_sha256"]
    )

    nb07 = decode(nb07_raw)

    require(
        nb07.get("acceptance_status") == "PASS"
        and nb07["artifacts"]["unified_documents"]["sha256"] ==
        c6["nb07_canonical_documents_file_sha256"]
        and nb07["artifacts"]["document_schema"]["sha256"] ==
        c6["nb07_document_schema_sha256"],
        "NB07_REFERENCE_LINEAGE_CHANGED"
    )

    checked_file(
        nb07["artifacts"]["document_schema"]["path"],
        c6["nb07_document_schema_sha256"]
    )

    checked_file(
        nb07["artifacts"]["unified_documents"]["path"],
        c6["nb07_canonical_documents_file_sha256"]
    )

    schema_refs = {
        "request": (
            "schemas/phase2/rag_request.schema.json",
            c4["schema_file_sha256"]
        ),
        "response": (
            "schemas/phase2/rag_response.schema.json",
            c5["response_schema_file_sha256"]
        ),
        "citation": (
            "schemas/phase2/rag_citation.schema.json",
            c6["citation_schema_sha256"]
        ),
        "policy": (
            c7["status_policy_path"],
            c7["status_policy_sha256"]
        ),
    }

    contracts = {}

    for label, (path, expected) in schema_refs.items():
        contracts[label] = decode(checked_file(path, expected))

        if label != "policy":
            Draft202012Validator.check_schema(contracts[label])

    require(
        contracts["request"]["properties"]["contract_version"]["const"]
        == "nb23.request.v1"
        and contracts["response"]["properties"]["contract_version"]["const"]
        == "nb23.response.v1"
        and contracts["policy"].get("live_runtime_status_mapping")
        == "NOT_VERIFIED",
        "DRAFT_CONTRACT_VERSIONS_OR_LIMITS_CHANGED"
    )

    request_v = Draft202012Validator(contracts["request"])
    response_v = Draft202012Validator(contracts["response"])
    citation_v = Draft202012Validator(contracts["citation"])
    policy = contracts["policy"]

    fixtures = decode(
        checked_file(
            c9["fixture_file"],
            c9["fixture_file_sha256"]
        )
    )

    samples = decode(
        checked_file(
            c11["sample_path"],
            c11["sample_sha256"]
        )
    )

    guide_bytes = checked_file(
        c11["guide_path"],
        c11["guide_sha256"]
    )

    require(
        guide_bytes.startswith(b'<div dir="rtl" align="right">')
        and guide_bytes.rstrip().endswith(b'</div>'),
        "C011_GUIDE_RTL_WRAPPER_INVALID"
    )

    require(
        fixtures.get("source_fingerprint") == fp
        and fixtures.get("origin") ==
        "SYNTHETIC_CONTRACT_FIXTURES_ONLY"
        and len(fixtures.get("fixture_cases", [])) == 41
        and samples.get("source_fingerprint") == fp
        and samples.get("fixture_origin") ==
        "SYNTHETIC_CONTRACT_FIXTURES_ONLY"
        and len(samples.get("conversations", [])) == 12
        and sum(
            len(c["turns"])
            for c in samples["conversations"]
        ) == 13,
        "C009_C011_EXPORTED_CENSUS_CHANGED"
    )

    positive = {
        x["name"]: x["bundle"]
        for x in fixtures["fixture_cases"]
        if x["expected_valid"] is True
    }

    require(
        len(positive) == 15
        and len(c6["sampled_source_roles"]) == 6,
        "C009_SOURCE_FIXTURE_CENSUS_CHANGED"
    )

    catalog = {}

    for role in c6["sampled_source_roles"]:
        key = "positive_source_" + role

        require(
            key in positive
            and len(positive[key]["resolved_citations"]) == 1,
            "SOURCE_ROLE_FIXTURE_MISSING:" + role
        )

        detail = positive[key]["resolved_citations"][0]

        require(
            citation_v.is_valid(detail),
            "SOURCE_ROLE_CITATION_INVALID:" + role
        )

        cid = detail["citation_id"]

        require(
            cid not in catalog,
            "DUPLICATE_SYNTHETIC_CITATION_ID"
        )

        catalog[cid] = detail

    # Independently recheck C009 and C011 from disk, not notebook-memory tests.
    def validate_bundle(bundle):
        if not request_v.is_valid(bundle["request"]):
            return False, "INVALID_REQUEST_SCHEMA"

        response = bundle["response"]

        if not response_v.is_valid(response):
            return False, "INVALID_RESPONSE_SCHEMA"

        if (
            response["source_fingerprint"] != fp
            or response["version_trace"] != {
                "source_fingerprint": fp,
                "adapter_version": identity["adapter_version"],
                "release_id": release_id,
            }
        ):
            return False, "VERSION_LOCK_MISMATCH"

        rule = policy["states"][response["outcome"]]
        answer_text = response["answer_text"]

        if (
            rule["answer_text"] == "REQUIRED"
            and not (
                isinstance(answer_text, str)
                and answer_text.strip()
            )
        ):
            return False, "REQUIRED_ANSWER_TEXT_ABSENT"

        if (
            rule["answer_text"] == "NULL"
            and answer_text is not None
        ):
            return False, "TEXT_PRESENT_ON_ERROR"

        if (
            rule["claims_and_citations"] == "LINKED_NONEMPTY"
            and not (
                response["claims"]
                and response["citations"]
            )
        ):
            return False, "ANSWER_REFERENCES_MISSING"

        if (
            rule["claims_and_citations"] == "EMPTY"
            and (
                response["claims"]
                or response["citations"]
            )
        ):
            return False, "FORBIDDEN_REFERENCES"

        if (
            rule["error"] == "NULL"
            and response["error"] is not None
        ):
            return False, "UNEXPECTED_ERROR_OBJECT"

        if (
            rule["error"] == "REQUIRED"
            and (
                response["error"] is None
                or response["error"]["code"]
                not in policy["error_codes_draft"]
            )
        ):
            return False, "UNKNOWN_RUNTIME_ERROR"

        if (
            response["outcome"] == "ERROR"
            and response["confidence"] is not None
        ):
            return False, "ERROR_CONFIDENCE_FORBIDDEN"

        claims = {
            x["claim_id"]: x
            for x in response["claims"]
        }

        citations = {
            x["citation_id"]: x
            for x in response["citations"]
        }

        if (
            len(claims) != len(response["claims"])
            or len(citations) != len(response["citations"])
        ):
            return False, "DUPLICATE_CLAIM_OR_CITATION_ID"

        for claim in response["claims"]:
            if not claim["citation_ids"]:
                return False, "UNCITED_CLAIM"

            for cid in claim["citation_ids"]:
                if (
                    cid not in citations
                    or claim["claim_id"]
                    not in citations[cid]["claim_ids"]
                ):
                    return False, "BROKEN_CLAIM_TO_CITATION"

        for citation in response["citations"]:
            for claim_id in citation["claim_ids"]:
                if (
                    claim_id not in claims
                    or citation["citation_id"]
                    not in claims[claim_id]["citation_ids"]
                ):
                    return False, "BROKEN_CITATION_TO_CLAIM"

        details = bundle["resolved_citations"]

        if len(details) != len(citations):
            return False, "CITATION_CARDINALITY_MISMATCH"

        seen = set()

        for detail in details:
            if not citation_v.is_valid(detail):
                return False, "CITATION_SCHEMA_INVALID"

            cid = detail["citation_id"]

            if (
                cid in seen
                or cid not in citations
                or detail["claim_ids"] != citations[cid]["claim_ids"]
            ):
                return False, "DETAILED_CITATION_LINK_MISMATCH"

            seen.add(cid)

            if (
                detail["source_type"] == "transcript"
                and detail["locator"]["source_line_start"]
                > detail["locator"]["source_line_end"]
            ):
                return False, "TRANSCRIPT_LINE_RANGE_INVERTED"

            if (
                cid not in catalog
                or detail != catalog[cid]
            ):
                return False, "SYNTHETIC_CITATION_CATALOG_MISMATCH"

        return True, "SYNTHETIC_CONTRACT_ONLY"

    independent_tests = []

    for row in fixtures["fixture_cases"]:
        valid, reason = validate_bundle(row["bundle"])

        require(
            valid is row["expected_valid"],
            "C009_INDEPENDENT_FIXTURE_FAILED:"
            + row["name"] + ":" + reason
        )

        independent_tests.append({
            "case": row["name"],
            "expected": row["expected_valid"],
            "observed": valid,
            "status": "PASS"
        })

    for conversation in samples["conversations"]:
        for index, turn in enumerate(conversation["turns"], 1):
            require(
                turn["origin"] == "SYNTHETIC_CONTRACT_FIXTURE"
                and turn["run_kind"] == "NOT_EXECUTED",
                "SAMPLE_PRETENDS_TO_BE_LIVE"
            )

            valid, reason = validate_bundle(turn["bundle"])

            require(
                valid,
                "C011_INDEPENDENT_TURN_FAILED:"
                + conversation["case_id"]
                + ":" + str(index) + ":" + reason
            )

            independent_tests.append({
                "case": (
                    conversation["case_id"]
                    + ":turn:" + str(index)
                ),
                "expected": True,
                "observed": valid,
                "status": "PASS"
            })

    require(
        len(independent_tests) == 54
        and all(
            x["status"] == "PASS"
            for x in independent_tests
        ),
        "INDEPENDENT_CONTRACT_TEST_COUNT_CHANGED"
    )

    # Evidence index does not include its own output hash.
    evidence_index = {
        "schema_version": "NB23_C012_NORMALIZED_EVIDENCE_INDEX_V2",
        "action": 10,
        "source_identity": identity,
        "source_fingerprint": fp,
        "normalized_source_contract_version":
            NB23_INPUT["contract_version"],
        "release_id_from_normalized_input": release_id,
        "upstream_release_accepted": accepted,
        "source_acceptance_evidence_verified_by_adapter":
            acceptance_proof,
        "read_only_inputs": sorted(
            verified.values(),
            key=lambda x: x["path"]
        ),
        "parent_receipt_paths": paths,
        "parent_receipt_sha256": hashes,
        "source_input_roles": role_paths,
        "contract_test_rechecks": {
            "C009_fixture_cases": 41,
            "C011_sample_turns": 13,
            "independent_cases_passed": len(independent_tests),
            "independent_cases_failed": 0,
        },
        "manifest_pointer_count_observed_in_C008":
            lock.get("checkpoint_manifest_pointer_count"),
        "real_producer_runtime_tested_here": False,
        "selected_release_corpus_equivalence_verified": False,
        "new_api_calls": 0,
        "git_operations": 0,
    }

    evidence_bytes = canon(evidence_index)
    evidence_sha = sha(evidence_bytes)

    evidence_path = work / (
        "NB23.C012.normalized_evidence_index."
        + evidence_sha + ".json"
    )

    blockers = [
        {
            "id": "B01",
            "status": "BLOCKED" if not accepted else "UPSTREAM_ONLY",
            "reason": (
                "PHASE2_RELEASE_NOT_ACCEPTED"
                if not accepted else
                "UPSTREAM_ACCEPTANCE_NOT_SUFFICIENT_FOR_PHASE3_HANDOFF"
            )
        },
        {
            "id": "B02",
            "status": (
                "BLOCKED" if unresolved
                else "NO_C008_GAP_REPORTED"
            ),
            "reason": "AUTHORITATIVE_COMPONENT_VERSION_GAPS",
            "components": unresolved
        },
        {
            "id": "B03",
            "status": "NOT_VERIFIED",
            "reason": "LIVE_PRODUCER_REQUEST_RESPONSE_STATUS_ADAPTER"
        },
        {
            "id": "B04",
            "status": "NOT_VERIFIED",
            "reason": "CHUNK_EVIDENCE_AND_SEMANTIC_CITATION"
        },
        {
            "id": "B05",
            "status": "NOT_VERIFIED",
            "reason": "RUNTIME_SAFETY_CONFIDENCE_PROVIDER_ERROR_TRANSPORT"
        },
        {
            "id": "B06",
            "status": "NOT_ESTABLISHED",
            "reason": "PREVIOUS_CONSUMER_SCHEMA_BACKWARD_COMPATIBILITY"
        },
        {
            "id": "B07",
            "status": "DEFERRED",
            "reason": "CLEAN_KERNEL_RUN_ALL_AND_END_TO_END_SMOKE_NOT_PROVEN_BY_THIS_CELL"
        },
    ]

    action_scopes = [
        (1, "C003", "SOURCE_BOUND_DRAFT_OR_UPSTREAM_ACCEPTED_ONLY"),
        (2, "C004", "DRAFT_REQUEST_SCHEMA_TESTED"),
        (3, "C005", "DRAFT_RESPONSE_SCHEMA_TESTED"),
        (4, "C006", "DOCUMENT_CITATION_WITH_LIMITATIONS"),
        (5, "C007", "DRAFT_STATUS_POLICY_TESTED"),
        (6, "C008_report", "SOURCE_LOCK_WITH_RECORDED_GAPS"),
        (7, "C009", "SYNTHETIC_CONTRACT_FIXTURES"),
        (8, "C010_report", "HYPOTHETICAL_DRIFT_NOT_REAL_COMPATIBILITY"),
        (9, "C011", "SYNTHETIC_SAMPLES_AND_GUIDE"),
        (10, "C012", "DRAFT_READINESS_AUDIT_NOT_FORMAL_ACCEPTANCE"),
    ]

    gate = {
        "schema_version": "NB23_C012_NORMALIZED_READINESS_GATE_V2",
        "action": 10,
        "source_fingerprint": fp,
        "source_ref": selected,
        "source_adapter_version": identity["adapter_version"],
        "normalized_source_contract_version":
            NB23_INPUT["contract_version"],
        "source_version_lock_sha256": hashes["C008_lock"],
        "evidence_index_path":
            evidence_path.relative_to(root).as_posix(),
        "evidence_index_file_sha256": evidence_sha,
        "action_ledger": [
            {
                "action": n,
                "receipt": key,
                "scope": scope
            }
            for n, key, scope in action_scopes
        ],
        "actions_accounted": 10,
        "technical_contract_integrity":
            "PASS_IN_SYNTHETIC_DRAFT_SCOPE",
        "independent_fixture_regression": {
            "passed": 54,
            "failed": 0
        },
        "upstream_release_accepted": accepted,
        "upstream_acceptance_evidence_verified_by_adapter":
            acceptance_proof,
        "upstream_release_gate_status": gate_status,
        "release_acceptance_gate": (
            "UPSTREAM_ACCEPTED_ONLY"
            if accepted else "BLOCKED_NOT_ACCEPTED"
        ),
        "component_version_gate": (
            "BLOCKED_UNRESOLVED_COMPONENTS"
            if unresolved else "C008_NO_UNRESOLVED_COMPONENTS"
        ),
        "authoritative_components_unresolved": unresolved,
        "runtime_contract_and_safety_gate": "NOT_VERIFIED",
        "citation_semantic_gate": "NOT_VERIFIED",
        "backward_compatibility_gate":
            "NO_VERIFIED_PREVIOUS_CONSUMER_SCHEMA",
        "end_to_end_smoke_gate": "NOT_EXECUTED",
        "clean_run_all_gate":
            "NOT_PROVEN_BY_C012_CELL_EXECUTION",
        "formal_readiness": "BLOCKED",
        "readiness_decision":
            "DRAFT_PREPARATION_COMPLETE_FORMAL_PHASE3_HANDOFF_NOT_READY",
        "formal_release_id": (
            release_id
            if accepted and acceptance_proof
            else None
        ),
        "accepted_for_phase3": False,
        "phase2_completed": False,
        "phase3_module00_accepted": False,
        "blocking_and_deferred_items": blockers,
        "module00_consumer_responsibilities": [
            "ACQUIRE_AND_VERIFY_ACCEPTED_PHASE2_RELEASE_MANIFEST",
            "VERIFY_DATA_CHUNK_EMBEDDING_INDEX_CONFIG_AND_MODEL_LOCK",
            "VERIFY_LIVE_REQUEST_RESPONSE_STATUS_AND_PRODUCER_ADAPTER",
            "VERIFY_WEBSITE_BOOK_TRANSCRIPT_LOCATORS_AND_ATTRIBUTION",
            "RUN_SEMANTIC_CITATION_AND_SAFETY_TESTS",
            "RUN_CLEAN_RESTART_RUN_ALL_AND_END_TO_END_SMOKE",
            "PROVE_BACKWARD_COMPATIBILITY_IF_PREVIOUS_CONSUMER_EXISTS",
            "RECORD_FORMAL_GATE_AND_ROLLBACK_TARGET",
        ],
        "no_formal_release_artifact_path_created": True,
        "new_api_calls": 0,
        "git_operations": 0,
    }

    gate_bytes = canon(gate)
    gate_sha = sha(gate_bytes)

    gate_path = work / (
        "NB23.C012.normalized_readiness_gate."
        + gate_sha + ".json"
    )

    release_text = (
        "پذیرش بالادستی در Adapter ثبت شده، اما تحویل فاز سوم هنوز پذیرفته نیست."
        if accepted else
        "Release بالادستی هنوز پذیرفته نشده است."
    )

    report_lines = [
        '<div dir="rtl" align="right">',
        '### NB23.C012 — ممیزی نسخه‌خنثی Draft و آمادگی تحویل فاز سوم',
        '',
        '**نتیجه:** قراردادهای Draft آزموده شده‌اند؛ تحویل رسمی فاز سوم **BLOCKED** است.',
        '',
        '**اثر انگشت نسخه:** `' + fp + '`',
        '',
        '**Release:** ' + release_text,
        '',
        '- شناسهٔ Release طبق ورودی نرمال‌شده: `'
        + (release_id or 'null') + '`.',
        '- Adapter: `' + identity["adapter_version"] + '`.',
        '',
        '#### شواهد فنی',
        '',
        '- ۱۱ رسید مستقل C003 تا C011 با SHA-256 و مسیر دقیق بازخوانی شد.',
        '- پنج گروه ضروری ورودی و قراردادهای Request/Response/Citation/Status بررسی شدند.',
        '- ۴۱ Fixture و ۱۳ Turn ذخیره‌شده، مستقل و آفلاین بررسی شدند: **۵۴ PASS**.',
        '- همهٔ نمونه‌های فوق مصنوعی‌اند و اجرای مدل واقعی یا کیفیت معنایی را اثبات نمی‌کنند.',
        '',
        '#### موانع پذیرش',
        '',
        '- وضعیت Gate بالادستی: `' + str(gate_status) + '`.',
        '- گروه‌های authoritative حل‌نشده: ' + str(len(unresolved)) + '.',
        '- Runtime، Status واقعی، Safety، Citation معنایی، Compatibility واقعی و Smoke مستقل تأیید نشده‌اند.',
        '- اجرای مستقل Restart Kernel → Run All هنوز با اجرای این سل اثبات نمی‌شود.',
        '',
        '#### اقدام بعدی',
        '',
        '- ثبت Adapter معتبر برای هر Schema جدید فقط در C002؛ سایر سل‌ها تنها NB23_INPUT را بخوانند.',
        '- آزمون A → B → A با ورودی واقعی و ثبت خروجی‌های جداگانه قبل از ادعای سوییچ کامل.',
        '- اجرای پاک Run All و Gateهای رسمی پس از آماده‌شدن Release معتبر.',
        '',
        '#### مراجع',
        '',
        '- Evidence Index: `'
        + evidence_path.relative_to(root).as_posix() + '`',
        '- Evidence SHA-256: `' + evidence_sha + '`',
        '- Readiness Gate: `'
        + gate_path.relative_to(root).as_posix() + '`',
        '- Gate SHA-256: `' + gate_sha + '`',
        '',
        '**وضعیت نهایی:** `DRAFT_PREPARATION_COMPLETE_FORMAL_PHASE3_HANDOFF_NOT_READY`',
        '',
        '</div>',
    ]

    report_md = "\n".join(report_lines).encode("utf-8")
    report_sha = sha(report_md)

    report_path = (
        root
        / "reports/phase2/release/nb23_draft_readiness"
        / fp
        / (
            "NB23.C012.normalized_draft_readiness."
            + report_sha + ".md"
        )
    )

    secret = re.compile(
        rb"(?:sk-[A-Za-z0-9_-]{12,}|ghp_[A-Za-z0-9]{12,}|"
        rb"AKIA[A-Z0-9]{16}|"
        rb"-----BEGIN (?:RSA |EC |OPENSSH )?PRIVATE KEY-----|"
        rb"[Bb]earer\s+[A-Za-z0-9_.-]{8,}|"
        rb"/Users/|/home/|C:\\Users\\)"
    )

    require(
        all(
            secret.search(raw) is None
            for raw in (evidence_bytes, gate_bytes, report_md)
        ),
        "SENSITIVE_PATTERN_FOUND_IN_OUTPUT"
    )

    require(
        report_md.startswith(b'<div dir="rtl" align="right">')
        and report_md.rstrip().endswith(b'</div>'),
        "FINAL_PERSIAN_MARKDOWN_RTL_INVALID"
    )

    immutable(evidence_path, evidence_bytes)
    immutable(gate_path, gate_bytes)
    immutable(report_path, report_md)

    require(
        decode(evidence_path.read_bytes()) == evidence_index
        and decode(gate_path.read_bytes()) == gate
        and report_path.read_bytes() == report_md
        and sha(evidence_path.read_bytes()) == evidence_sha
        and sha(gate_path.read_bytes()) == gate_sha
        and sha(report_path.read_bytes()) == report_sha,
        "C012_INDEPENDENT_RELOAD_FAILED"
    )

    return {
        "status":
            "NB23_C012_NORMALIZED_DRAFT_PREPARATION_COMPLETE_FORMAL_HANDOFF_BLOCKED",
        "action": 10,
        "source_fingerprint": fp,
        "source_adapter_version": identity["adapter_version"],
        "release_id_from_normalized_input": release_id,
        "upstream_release_accepted": accepted,
        "actions_accounted": 10,
        "verified_input_files": len(verified),
        "independent_contract_tests": len(independent_tests),
        "independent_tests_passed": len(independent_tests),
        "upstream_release_gate": gate_status,
        "formal_readiness": "BLOCKED",
        "authoritative_component_groups_unresolved": len(unresolved),
        "evidence_index_path":
            evidence_path.relative_to(root).as_posix(),
        "evidence_index_sha256": evidence_sha,
        "readiness_gate_path":
            gate_path.relative_to(root).as_posix(),
        "readiness_gate_sha256": gate_sha,
        "draft_report_path":
            report_path.relative_to(root).as_posix(),
        "draft_report_sha256": report_sha,
        "reload_validation": "PASS",
        "accepted_for_phase3": False,
        "phase2_completed": False,
        "new_api_calls": 0,
        "git_operations": 0,
    }


nb23_c012_result = nb23_c012_final_readiness()

print(json.dumps(
    nb23_c012_result,
    ensure_ascii=False,
    indent=2
))
```
