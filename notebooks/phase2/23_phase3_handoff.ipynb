```python
# NB23.C001 — Central imports and bootstrap
# Scope: standard-library imports only; no API, filesystem writes, or Git operations.

import os
import json
import hashlib

from pathlib import Path
from datetime import datetime, timezone
from collections import Counter
```

```python
# NB23.C002 — Action 01, package 1/2: read-only source/version preflight.
# No writes, API, Git operations, or acceptance changes. Only NB23_SOURCE_REF selects source version.
NB23_SOURCE_REF = (
    "reports/phase2/release/NB22.C054.quality_disposition/"
    "C054.supplemented_NB23_handoff."
    "f41ed8f332785201ec29a7e9ca6c7f60c122730044419d5b833745fca17c059f.json"
)


def nb23_c002_preflight():
    def sha256(data):
        return hashlib.sha256(data).hexdigest()

    def no_duplicates(items):
        result = {}
        for key, value in items:
            if key in result:
                raise ValueError("DUPLICATE_JSON_KEY")
            result[key] = value
        return result

    report = {
        "cell": "NB23.C002",
        "action": 1,
        "package": "1_OF_2_SOURCE_PREFLIGHT",
        "source_ref": NB23_SOURCE_REF,
        "new_api_requests": 0,
        "files_written": 0,
        "git_changes": 0,
        "release_acceptance_granted": False,
    }

    markers = (
        Path("notebooks/phase2/checklist/NB23_phase3_handoff_checklist_fa.md"),
        Path("data/Road_Maps/RAG_finance_phase2_operational_roadmap.md"),
    )

    override = os.getenv("NB23_REPO_ROOT", "").strip()

    roots_to_check = (
        [Path(override).expanduser()] if override else []
    ) + [
        Path.cwd(),
        *Path.cwd().parents
    ]

    roots = {
        candidate.resolve()
        for candidate in roots_to_check
        if all((candidate / marker).is_file() for marker in markers)
    }

    if len(roots) != 1:
        report.update({
            "status": (
                "REPO_ROOT_NOT_FOUND"
                if not roots
                else "REPO_ROOT_AMBIGUOUS"
            ),
            "root_candidates_matching": len(roots),
            "root_override_variable": "NB23_REPO_ROOT",
            "expected_markers": [
                x.as_posix() for x in markers
            ],
        })
        return report

    root = roots.pop()
    report["repo_folder_name"] = root.name

    ref = Path(NB23_SOURCE_REF)

    if (
        not isinstance(NB23_SOURCE_REF, str)
        or not NB23_SOURCE_REF.strip()
        or ref.is_absolute()
        or ".." in ref.parts
        or ref.suffix.lower() != ".json"
    ):
        report["status"] = "INVALID_SOURCE_REF"
        return report

    source_file = (root / ref).resolve()

    if not source_file.is_relative_to(root):
        report["status"] = "SOURCE_REF_OUTSIDE_REPO"
        return report

    if not source_file.is_file():
        # Known alternative layout under notebooks/phase2.
        # Do not silently switch to another source.
        alternative = (
            root / "notebooks/phase2" / ref
        ).resolve()

        known_alternatives = []

        if (
            alternative.is_relative_to(root)
            and alternative.is_file()
        ):
            known_alternatives.append(
                alternative.relative_to(root).as_posix()
            )

        report.update({
            "status": "SOURCE_NOT_FOUND",
            "known_alternative_paths_if_present": known_alternatives,
            "source_available": False,
            "next_step": (
                "Find the real local Handoff path; "
                "change only NB23_SOURCE_REF"
            ),
        })
        return report

    size = source_file.stat().st_size

    if size <= 0 or size > 32 * 1024 * 1024:
        report.update({
            "status": "SOURCE_SIZE_OUT_OF_SCOPE",
            "bytes": size,
        })
        return report

    raw = source_file.read_bytes()

    report.update({
        "source_available": True,
        "source_bytes": len(raw),
        "source_raw_byte_sha256": sha256(raw),
        "filename_hash_assumed_to_be_raw_hash": False,
    })

    try:
        obj = json.loads(
            raw.decode("utf-8-sig"),
            object_pairs_hook=no_duplicates,
            parse_constant=lambda s: (
                _ for _ in ()
            ).throw(
                ValueError("INVALID_JSON_CONSTANT")
            ),
        )

        if not isinstance(obj, dict):
            raise ValueError("SOURCE_ROOT_NOT_OBJECT")

        normalized = json.dumps(
            obj,
            ensure_ascii=False,
            sort_keys=True,
            separators=(",", ":"),
            allow_nan=False,
        ).encode("utf-8")

    except (
        UnicodeDecodeError,
        ValueError,
        TypeError,
    ) as exc:
        report.update({
            "status": "SOURCE_JSON_INVALID",
            "error_type": type(exc).__name__,
            "error_code": str(exc)[:100],
        })
        return report

    identity_fields = (
        "schema_version",
        "artifact_type",
        "report_type",
        "release_id",
        "release_status",
        "acceptance_status",
        "acceptance_decision",
        "accepted_for_phase3",
        "status",
    )

    direct_metadata = {}

    for key in identity_fields:
        value = obj.get(key)

        if isinstance(
            value,
            (type(None), bool, int, float),
        ):
            direct_metadata[key] = value

        elif (
            isinstance(value, str)
            and len(value) <= 120
        ):
            direct_metadata[key] = value

    relevant_nested = {}

    for key, value in obj.items():
        if (
            isinstance(value, dict)
            and any(
                term in key.lower()
                for term in (
                    "manifest",
                    "release",
                    "accept",
                    "handoff",
                    "source",
                    "input",
                    "gate",
                )
            )
        ):
            relevant_nested[key] = sorted(
                value.keys()
            )[:30]

    report.update({
        "status": "SOURCE_READ_AND_HASHED",
        "source_json_canonical_sha256":
            sha256(normalized),
        "source_top_level_key_count": len(obj),
        "source_top_level_keys":
            sorted(obj)[:90],
        "source_field_types": {
            k: type(v).__name__
            for k, v in list(obj.items())[:60]
        },
        "direct_metadata_as_recorded":
            direct_metadata,
        "nested_identity_field_names":
            dict(list(relevant_nested.items())[:12]),
        "adapter_state":
            "AWAITING_SCHEMA_SPECIFIC_MAPPING",
        "acceptance_evidence_verified": False,
        "handoff_mode":
            "DRAFT_ONLY_UNTIL_GATE_PROVEN",
        "next_step": (
            "Action 01 package 2: bind actual manifest "
            "and acceptance with source-version-safe adapter"
        ),
    })

    return report


nb23_c002_result = nb23_c002_preflight()

print(json.dumps(
    nb23_c002_result,
    ensure_ascii=False,
    indent=2,
))
```

```python
# NB23.C003 — Action 01, package 2/2: versioned source adapter and evidence binding.
# Scope: local read-only upstream inspection; one immutable NB23 receipt; no API or Git actions.
import os
import re
import json
import hashlib
import tempfile
from pathlib import Path


def nb23_c003_bind_source():
    def require(condition, code):
        if not condition:
            raise RuntimeError(code)

    def hash_bytes(data):
        return hashlib.sha256(data).hexdigest()

    def canonical(value):
        return json.dumps(value, ensure_ascii=False, sort_keys=True,
                          separators=(",", ":"), allow_nan=False).encode("utf-8")

    def distinct(pairs):
        result = {}
        for name, value in pairs:
            if name in result:
                raise ValueError("DUPLICATE_JSON_KEY:" + name)
            result[name] = value
        return result

    def read_json(data):
        return json.loads(data.decode("utf-8-sig"), object_pairs_hook=distinct,
                          parse_constant=lambda value: (_ for _ in ()).throw(
                              ValueError("INVALID_JSON_CONSTANT:" + value)))

    def in_repo(path, root):
        return path == root or root in path.parents

    def small_scalar(v):
        return v if type(v) in (bool, int, float, type(None)) or (
            isinstance(v, str) and len(v) <= 150) else "NON_SCALAR_OR_LONG"

    markers = (
        Path("notebooks/phase2/checklist/NB23_phase3_handoff_checklist_fa.md"),
        Path("data/Road_Maps/RAG_finance_phase2_operational_roadmap.md"),
    )
    if "NB23_SOURCE_REF" not in globals():
        raise RuntimeError("RUN_NB23_C002_FIRST_SOURCE_REF_NOT_SET")
    selected = globals()["NB23_SOURCE_REF"]
    require(isinstance(selected, str) and selected.strip(), "SOURCE_REF_INVALID")
    ref = Path(selected)
    require(not ref.is_absolute() and ".." not in ref.parts and ref.suffix == ".json",
            "SOURCE_REF_MUST_BE_REPO_RELATIVE_JSON")
    override = os.environ.get("NB23_REPO_ROOT", "").strip()
    start = Path.cwd()
    options = ([Path(override).expanduser()] if override else []) + [start, *start.parents]
    roots = {p.resolve() for p in options if all((p / m).is_file() for m in markers)}
    require(len(roots) == 1, "ROOT_MISSING_OR_AMBIGUOUS")
    root = roots.pop()
    source = (root / ref).resolve()
    require(in_repo(source, root) and source.is_file(), "SELECTED_SOURCE_NOT_A_FILE_IN_REPO")
    require(0 < source.stat().st_size <= 32 * 1024 * 1024, "SOURCE_SIZE_INVALID")
    source_raw = source.read_bytes()
    source_obj = read_json(source_raw)
    require(isinstance(source_obj, dict), "SOURCE_JSON_NOT_OBJECT")
    source_sha = hash_bytes(source_raw)
    source_canonical_sha = hash_bytes(canonical(source_obj))
    recorded_name_sha = re.search(r"\.([a-f0-9]{64})\.json$", source.name)
    filename_match = (source_sha == recorded_name_sha.group(1)) if recorded_name_sha else None

    # The adapter registry is explicit. An unknown future NB22B schema stops safely here.
    adapter_key = (source_obj.get("schema_version"), source_obj.get("artifact_type"))
    supported = ("NB22_C054_HANDOFF_QUALITY_SUPPLEMENT_V1",
                 "NB22_TO_NB23_EVIDENCE_STATUS_HANDOFF")
    if adapter_key != supported:
        return {
            "status": "ADAPTER_REQUIRED_UNRECOGNIZED_SOURCE_SCHEMA",
            "action": 1, "source_ref": selected,
            "observed_schema_version": small_scalar(adapter_key[0]),
            "observed_artifact_type": small_scalar(adapter_key[1]),
            "source_raw_byte_sha256": source_sha,
            "new_api_calls": 0, "files_written": 0,
            "release_acceptance_granted": False,
        }
    require(filename_match is not False, "C054_FILENAME_AND_BYTES_SHA256_CONFLICT")
    require(isinstance(source_obj.get("inputs"), dict), "C054_INPUTS_NOT_OBJECT")
    require(isinstance(source_obj.get("parent_handoff"), dict), "C054_PARENT_NOT_OBJECT")
    expected_inputs = ("checkpoint_disposition", "checkpoint_manifest",
                       "checkpoint_validation", "release_gate")
    require(all(k in source_obj["inputs"] for k in expected_inputs),
            "C054_REQUIRED_INPUT_BINDINGS_MISSING")

    # The present C054 supplement records no accepted Release. Preserve that fact.
    acceptance_fields = (
        "release_acceptance_granted", "accepted_release_input_established",
        "action28_accepted_release_input_condition_met",
        "phase2_completion_claimed", "formal_NB23_start_authorized_by_this_handoff",
    )
    accepted_facts = {}
    for key in acceptance_fields:
        value = source_obj.get(key)
        require(type(value) is bool, "C054_BOOLEAN_CONTRACT_MISSING:" + key)
        accepted_facts[key] = value
    require(not any(accepted_facts.values()), "C054_ACCEPTANCE_CHANGE_NEEDS_NEW_GATE_REVIEW")

    raw_sha_names = ("file_bytes_sha256", "raw_byte_sha256", "raw_file_sha256",
                     "byte_sha256", "file_sha256")
    path_keys = ("path", "relative_path", "artifact_path", "file_path")
    sha_names = ("sha256", "artifact_sha256", "canonical_sha256",
                 "logical_sha256", "git_blob_sha", "manifest_sha256")

    def extract_pointers(node, branch="", depth=0):
        found = []
        if isinstance(node, str) and branch == "":
            if node.endswith(".json"):
                found.append(("$", node, {}))
        elif isinstance(node, dict) and depth <= 4:
            for key in path_keys:
                candidate = node.get(key)
                if isinstance(candidate, str) and candidate.strip():
                    found.append((branch + "." + key, candidate, node))
            for key, value in node.items():
                if isinstance(value, (dict, list)):
                    found += extract_pointers(value, branch + "." + key, depth + 1)
        elif isinstance(node, list) and depth <= 4:
            for i, item in enumerate(node[:30]):
                if isinstance(item, (dict, list)):
                    found += extract_pointers(item, branch + f"[{i}]", depth + 1)
        return found

    def check_record(group, entry):
        candidates = extract_pointers(entry)
        if not candidates:
            return [{"group": group, "status": "PATH_NOT_EXPLICITLY_LOCATED",
                     "pointer": None, "expected_hash_fields": []}]
        outcomes = []
        seen = set()
        for where, reported_path, container in candidates:
            if (where, reported_path) in seen:
                continue
            seen.add((where, reported_path))
            declared = {name: value for name in (*raw_sha_names, *sha_names)
                        if isinstance((value := container.get(name)), str)
                        and re.fullmatch(r"[a-fA-F0-9]{64}", value)}
            item = {"group": group, "pointer": where,
                    "expected_hash_fields": sorted(declared),
                    "path": reported_path if not Path(reported_path).is_absolute()
                            else "ABSOLUTE_PATH_NOT_EXPORTED"}
            p = Path(reported_path)
            if p.is_absolute() or ".." in p.parts:
                item["status"] = "EXTERNAL_OR_UNSAFE_PATH_UNVERIFIED"
            else:
                full = (root / p).resolve()
                if not in_repo(full, root):
                    item["status"] = "OUTSIDE_REPO_UNVERIFIED"
                elif not full.is_file():
                    item["status"] = "DECLARED_FILE_NOT_ACCESSIBLE"
                elif full.stat().st_size > 64 * 1024 * 1024:
                    item["status"] = "FILE_TOO_LARGE_FOR_THIS_BINDING_CHECK"
                else:
                    file_raw = full.read_bytes()
                    actual_sha = hash_bytes(file_raw)
                    item["actual_file_bytes_sha256"] = actual_sha
                    item["file_size_bytes"] = len(file_raw)
                    mismatches = [key for key in raw_sha_names
                                  if key in declared and declared[key].lower() != actual_sha]
                    if mismatches:
                        item["status"] = "DECLARED_BYTE_HASH_MISMATCH"
                        item["mismatched_byte_hash_fields"] = mismatches
                    else:
                        item["status"] = ("BYTE_HASH_VERIFIED" if
                                          any(key in declared for key in raw_sha_names)
                                          else "FILE_READ_HASHED_NO_TYPED_BYTE_HASH")
                    item["untyped_hash_comparisons"] = {
                        key: ("MATCHES_FILE_BYTES" if value.lower() == actual_sha
                              else "NOT_A_FILE_BYTES_MATCH_HASH_KIND_UNVERIFIED")
                        for key, value in declared.items() if key in sha_names
                    }
                    # Decisions are extracted as quotes/metadata, not evaluated as new gates.
                    if group == "release_gate" and full.suffix == ".json" and len(file_raw) < 8 * 1024 * 1024:
                        try:
                            gate = read_json(file_raw)
                            if isinstance(gate, dict):
                                item["observed_gate_fields"] = {
                                    key: small_scalar(gate[key]) for key in (
                                        "schema_version", "status", "gate_status", "overall_status",
                                        "acceptance_status", "release_accepted", "accepted_for_phase3")
                                    if key in gate
                                }
                        except (ValueError, UnicodeDecodeError):
                            item["gate_json_read_status"] = "INVALID_JSON_OR_DUPLICATE_KEYS"
            outcomes.append(item)
        return outcomes

    references = []
    for name in expected_inputs:
        references.extend(check_record(name, source_obj["inputs"][name]))
    references.extend(check_record("parent_handoff", source_obj["parent_handoff"]))
    unresolved = [r["group"] for r in references if r["status"] not in (
        "BYTE_HASH_VERIFIED", "FILE_READ_HASHED_NO_TYPED_BYTE_HASH")]
    fingerprint_input = {
        "source_ref": selected,
        "source_raw_byte_sha256": source_sha,
        "source_schema_version": supported[0],
        "source_artifact_type": supported[1],
        "adapter_version": "NB23_C054_SOURCE_ADAPTER_V1",
    }
    fingerprint = hash_bytes(canonical(fingerprint_input))
    payload = {
        "schema_version": "NB23_C003_SOURCE_BINDING_RECEIPT_V1",
        "action": 1, "action_scope": "DRAFT_INPUT_BINDING_ONLY",
        "source_identity": fingerprint_input,
        "source_fingerprint": fingerprint,
        "source_json_canonical_sha256": source_canonical_sha,
        "source_filename_hash_matches_bytes": filename_match,
        "input_bindings": references,
        "accepted_release_facts_from_C054": accepted_facts,
        "C054_status_as_recorded": source_obj.get("status"),
        "unresolved_binding_groups": sorted(set(unresolved)),
        "formal_release_accepted": False,
        "formal_NB23_handoff_ready": False,
        "adapter_registry_status": "C054_SUPPORTED_FUTURE_SCHEMA_REQUIRES_ADAPTER",
        "input_evidence_integrity": ("PARTIAL_OR_HELD" if unresolved else "ALL_DECLARED_POINTERS_READABLE"),
        "new_api_calls": 0,
        "source_files_modified": 0,
        "git_operations": 0,
    }
    encoded = canonical(payload)
    receipt_sha = hash_bytes(encoded)
    output_dir = root / "artifacts/phase2/phase3_handoff/work" / fingerprint
    output_dir.mkdir(parents=True, exist_ok=True)
    destination = output_dir / ("NB23.C003.source_binding." + receipt_sha + ".json")
    fd, temp = tempfile.mkstemp(dir=output_dir, prefix=".NB23.C003.")
    try:
        with os.fdopen(fd, "wb") as stream:
            stream.write(encoded)
            stream.flush()
            os.fsync(stream.fileno())
        try:
            os.link(temp, destination)
        except FileExistsError:
            require(destination.read_bytes() == encoded, "IMMUTABLE_RECEIPT_COLLISION")
    finally:
        if os.path.exists(temp):
            os.unlink(temp)
    require(hash_bytes(destination.read_bytes()) == receipt_sha
            and read_json(destination.read_bytes()) == payload,
            "WRITTEN_RECEIPT_FAILED_RELOAD")
    return {
        "status": ("NB23_C003_DRAFT_SOURCE_BOUND" if not unresolved
                   else "NB23_C003_DRAFT_WITH_UNRESOLVED_BINDINGS"),
        "action": 1, "package": "2_OF_2",
        "source_fingerprint": fingerprint,
        "source_raw_byte_sha256": source_sha,
        "adapter_version": fingerprint_input["adapter_version"],
        "reference_binding_counts": {
            "total": len(references),
            "typed_byte_hash_verified": sum(x["status"] == "BYTE_HASH_VERIFIED" for x in references),
            "readable_without_typed_byte_hash": sum(x["status"] == "FILE_READ_HASHED_NO_TYPED_BYTE_HASH" for x in references),
            "unresolved": len(unresolved),
        },
        "unresolved_binding_groups": sorted(set(unresolved)),
        "reference_bindings": references,
        "release_accepted": False,
        "handoff_mode": "DRAFT_ONLY",
        "new_api_calls": 0, "git_operations": 0,
        "artifact_path": destination.relative_to(root).as_posix(),
        "artifact_sha256": receipt_sha,
        "reload_validation": "PASS",
        "action_01_formal_acceptance_complete": False,
    }


nb23_c003_result = nb23_c003_bind_source()
print(json.dumps(nb23_c003_result, ensure_ascii=False, indent=2))
```

```python
# NB23.C007 — Action 05: source-bound status/error contract, cross-field rules and immutable tests.
# No API, model inference, Git modification, or overwrite of earlier artifacts.
import os
import ast
import re
import json
import copy
import hashlib
import tempfile
from pathlib import Path
from jsonschema import Draft202012Validator


def nb23_c007_status_error_contract():
    def ensure(ok, code):
        if not ok:
            raise RuntimeError(code)

    def sha(data):
        return hashlib.sha256(data).hexdigest()

    def canonical(obj):
        return json.dumps(
            obj, ensure_ascii=False, sort_keys=True,
            separators=(",", ":"), allow_nan=False
        ).encode("utf-8")

    def unique_pairs(pairs):
        obj = {}
        for key, value in pairs:
            ensure(key not in obj, "DUPLICATE_JSON_KEY:" + key)
            obj[key] = value
        return obj

    def decode(raw):
        return json.loads(
            raw.decode("utf-8-sig"),
            object_pairs_hook=unique_pairs,
            parse_constant=lambda s: (
                _ for _ in ()
            ).throw(ValueError(s))
        )

    def get_receipt(folder, prefix):
        files = sorted(folder.glob(prefix + ".*.json"))
        ensure(
            len(files) == 1,
            "DEPENDENCY_RECEIPT_MISSING_OR_DUPLICATE:" + prefix
        )
        path = files[0]
        match = re.fullmatch(
            re.escape(prefix) + r"\.([0-9a-f]{64})\.json",
            path.name
        )
        ensure(
            match is not None,
            "INVALID_DEPENDENCY_RECEIPT_NAME:" + prefix
        )
        raw = path.read_bytes()
        ensure(
            sha(raw) == match.group(1),
            "DEPENDENCY_RECEIPT_HASH_INVALID:" + prefix
        )
        return decode(raw), sha(raw)

    def immutable(path, raw):
        path.parent.mkdir(parents=True, exist_ok=True)
        fd, temp = tempfile.mkstemp(
            dir=path.parent,
            prefix=".NB23.C007."
        )
        try:
            with os.fdopen(fd, "wb") as f:
                f.write(raw)
                f.flush()
                os.fsync(f.fileno())
            try:
                os.link(temp, path)
            except FileExistsError:
                ensure(
                    path.read_bytes() == raw,
                    "IMMUTABLE_PATH_CONFLICT:" + path.name
                )
        finally:
            if os.path.exists(temp):
                os.unlink(temp)
        ensure(
            path.read_bytes() == raw,
            "RELOAD_BYTES_INVALID:" + path.name
        )

    # Source selection and exact version binding.
    ensure(
        "NB23_SOURCE_REF" in globals(),
        "C002_SELECTION_MISSING"
    )
    selection = globals()["NB23_SOURCE_REF"]
    ensure(
        isinstance(selection, str) and selection.strip(),
        "INVALID_SOURCE_REF"
    )
    ref = Path(selection)
    ensure(
        not ref.is_absolute()
        and ".." not in ref.parts
        and ref.suffix == ".json",
        "INVALID_SOURCE_RELATIVE_PATH"
    )

    marker = Path(
        "notebooks/phase2/checklist/"
        "NB23_phase3_handoff_checklist_fa.md"
    )
    configured = os.getenv("NB23_REPO_ROOT", "").strip()
    locations = (
        [Path(configured).expanduser()] if configured else []
    ) + [
        Path.cwd(),
        *Path.cwd().parents
    ]
    roots = {
        p.resolve()
        for p in locations
        if (p / marker).is_file()
    }
    ensure(
        len(roots) == 1,
        "PROJECT_ROOT_MISSING_OR_AMBIGUOUS"
    )
    root = roots.pop()

    source_path = (root / ref).resolve()
    ensure(
        root in source_path.parents and source_path.is_file(),
        "SOURCE_NOT_AVAILABLE"
    )
    source_raw = source_path.read_bytes()
    source = decode(source_raw)
    ensure(
        isinstance(source, dict)
        and source.get("schema_version")
        == "NB22_C054_HANDOFF_QUALITY_SUPPLEMENT_V1"
        and source.get("artifact_type")
        == "NB22_TO_NB23_EVIDENCE_STATUS_HANDOFF",
        "UNRECOGNIZED_SOURCE_ADAPTER"
    )

    identity = {
        "source_ref": selection,
        "source_raw_byte_sha256": sha(source_raw),
        "source_schema_version": source["schema_version"],
        "source_artifact_type": source["artifact_type"],
        "adapter_version": "NB23_C054_SOURCE_ADAPTER_V1",
    }
    fp = sha(canonical(identity))
    work = (
        root
        / "artifacts/phase2/phase3_handoff/work"
        / fp
    )

    # Reuse the already verified results, not historical computations.
    c3, h3 = get_receipt(
        work, "NB23.C003.source_binding"
    )
    c4, h4 = get_receipt(
        work, "NB23.C004.request_contract_test"
    )
    c5, h5 = get_receipt(
        work, "NB23.C005.response_contract_test"
    )
    c6, h6 = get_receipt(
        work, "NB23.C006.citation_resolver_test"
    )

    ensure(
        c3.get("source_identity") == identity
        and c3.get("unresolved_binding_groups") == []
        and c3.get("formal_release_accepted") is False,
        "C003_BINDING_CONFLICT"
    )
    ensure(
        c4.get("source_fingerprint") == fp
        and c4.get("C003_receipt_sha256") == h3
        and c4.get("passed_fixtures") == 12
        and c4.get("release_accepted") is False,
        "C004_DEPENDENCY_CONFLICT"
    )
    ensure(
        c5.get("source_fingerprint") == fp
        and c5.get("C003_receipt_file_sha256") == h3
        and c5.get("C004_receipt_file_sha256") == h4
        and c5.get("fixture_pass_count") == 20
        and c5.get("release_accepted") is False,
        "C005_DEPENDENCY_CONFLICT"
    )
    ensure(
        c6.get("source_fingerprint") == fp
        and c6.get("input_receipt_sha256") == {
            "C003": h3,
            "C004": h4,
            "C005": h5
        }
        and c6.get("tests_passed") == 17
        and c6.get("tests_failed") == 0
        and c6.get("canonical_document_count") == 19456
        and c6.get("release_accepted") is False,
        "C006_DEPENDENCY_CONFLICT"
    )

    # Verify the exact schemas that previous actions tested.
    schemas = {}
    for label, path, expected in [
        (
            "request",
            "schemas/phase2/rag_request.schema.json",
            c4.get("schema_file_sha256")
        ),
        (
            "response",
            "schemas/phase2/rag_response.schema.json",
            c5.get("response_schema_file_sha256")
        ),
        (
            "citation",
            "schemas/phase2/rag_citation.schema.json",
            c6.get("citation_schema_sha256")
        ),
    ]:
        file = (root / path).resolve()
        ensure(
            root in file.parents and file.is_file(),
            "MISSING_CONTRACT_SCHEMA:" + label
        )
        raw = file.read_bytes()
        ensure(
            sha(raw) == expected,
            "CONTRACT_SCHEMA_BYTE_HASH_MISMATCH:" + label
        )
        schema = decode(raw)
        Draft202012Validator.check_schema(schema)
        schemas[label] = schema

    ensure(
        schemas["request"]["properties"][
            "contract_version"
        ]["const"] == "nb23.request.v1"
        and schemas["response"]["properties"][
            "contract_version"
        ]["const"] == "nb23.response.v1",
        "CONTRACT_VERSION_CHANGED"
    )
    response_validator = Draft202012Validator(
        schemas["response"]
    )

    actual_states = schemas["response"]["properties"][
        "outcome"
    ]["enum"]
    expected_states = [
        "ANSWER",
        "NO_ANSWER",
        "CLARIFICATION",
        "SAFETY_REFUSAL",
        "PARTIAL",
        "ERROR",
    ]
    ensure(
        actual_states == expected_states,
        "RESPONSE_STATUS_ENUM_CHANGED"
    )
    ensure(
        c6.get("chunk_evidence_lookup_tested") is False
        and c6.get("claim_semantic_entailment_tested") is False,
        "C006_SCOPE_UNEXPECTED"
    )

    # The historical generator's actual interface is text-only.
    generator_file = (
        root / c4["generator_reference"]["path"]
    )
    ensure(
        generator_file.is_file(),
        "HISTORICAL_GENERATOR_REFERENCE_MISSING"
    )
    generator_raw = generator_file.read_bytes()
    ensure(
        sha(generator_raw)
        == c4["generator_reference"]["file_bytes_sha256"],
        "GENERATOR_REFERENCE_CHANGED_SINCE_C004"
    )

    source_ast = ast.parse(
        generator_raw.decode("utf-8-sig")
    )
    observed_methods = [
        node
        for cls in source_ast.body
        if isinstance(cls, ast.ClassDef)
        and cls.name == "ArvanGenerator"
        for node in cls.body
        if isinstance(node, ast.FunctionDef)
        and node.name == "generate_answer"
    ]
    ensure(
        len(observed_methods) == 1
        and [
            x.arg for x in observed_methods[0].args.args
        ] == ["self", "question", "context"],
        "HISTORICAL_GENERATOR_SIGNATURE_CHANGED"
    )
    observed_return_annotation = observed_methods[0].returns
    ensure(
        isinstance(observed_return_annotation, ast.Name)
        and observed_return_annotation.id == "str",
        "HISTORICAL_GENERATOR_RETURN_TYPE_NOT_TEXT"
    )

    # Draft consumer error codes, not proven producer enums.
    error_codes = [
        "INPUT_VALIDATION_ERROR",
        "PROVIDER_REQUEST_FAILED",
        "PROVIDER_TIMEOUT",
        "PROVIDER_OUTCOME_UNKNOWN",
        "SOURCE_BINDING_ERROR",
        "CONTRACT_MISMATCH",
        "INTERNAL_ERROR",
    ]

    policy = {
        "schema_version": "NB23_STATUS_ERROR_POLICY_V1",
        "request_contract_version": "nb23.request.v1",
        "response_contract_version": "nb23.response.v1",
        "source_schema_fingerprint_independent": True,
        "status_names_are_consumer_draft_not_producer_enums": True,
        "automatic_text_to_status_inference_authorized": False,
        "producer_mapping": {
            "source_path": c4["generator_reference"]["path"],
            "source_file_bytes_sha256": sha(generator_raw),
            "observed_method": (
                "ArvanGenerator.generate_answer"
                "(self, question, context) -> str"
            ),
            "producer_status_enum": None,
            "mapping_status": "NOT_ESTABLISHED_TEXT_ONLY_INTERFACE",
            "all_consumer_statuses": {
                key: "UNMAPPED_NO_PROVEN_PRODUCER_ENUM"
                for key in actual_states
            },
        },
        "states": {
            "ANSWER": {
                "answer_text": "REQUIRED",
                "claims_and_citations": "LINKED_NONEMPTY",
                "error": "NULL",
                "classification": "DRAFT_PRESENTATION_ONLY",
            },
            "PARTIAL": {
                "answer_text": "REQUIRED",
                "claims_and_citations": "LINKED_NONEMPTY",
                "error": "NULL",
                "classification": "DRAFT_PRESENTATION_ONLY",
            },
            "NO_ANSWER": {
                "answer_text": "OPTIONAL_EXPLANATION",
                "claims_and_citations": "EMPTY",
                "error": "NULL",
                "classification": (
                    "REQUIRES_ACTUAL_ANSWERABILITY_DECISION"
                ),
            },
            "CLARIFICATION": {
                "answer_text": "REQUIRED",
                "claims_and_citations": "EMPTY",
                "error": "NULL",
                "classification": (
                    "REQUIRES_ACTUAL_CLARIFICATION_DECISION"
                ),
            },
            "SAFETY_REFUSAL": {
                "answer_text": "REQUIRED",
                "claims_and_citations": "EMPTY",
                "error": "NULL",
                "classification": "REQUIRES_ACTUAL_SAFETY_DECISION",
            },
            "ERROR": {
                "answer_text": "NULL",
                "claims_and_citations": "EMPTY",
                "error": "REQUIRED",
                "classification": "RUNTIME_ERROR_NOT_ABSTENTION",
            },
        },
        "error_codes_draft": error_codes,
        "confidence": {
            "missing": None,
            "zero_means": "MEASURED_ZERO_NOT_MISSING",
            "non_null_value_requires_measurement_provenance": True,
        },
        "unknown_provider_attempt_must_not_become_no_answer": True,
        "http_status_streaming_backend": "PHASE3_OUT_OF_SCOPE",
        "citation_entailment": "NOT_ASSESSED_IN_C007",
        "live_runtime_status_mapping": "NOT_VERIFIED",
    }

    ensure(
        set(policy["states"]) == set(expected_states)
        and len(policy["states"]) == len(expected_states),
        "STATUS_POLICY_NOT_EXHAUSTIVE"
    )
    ensure(
        len(error_codes) == len(set(error_codes)),
        "DUPLICATE_ERROR_CODE"
    )

    # Cross-field checks supplement, but do not change, C005 JSON Schema.
    def validate_draft_response(response):
        if not response_validator.is_valid(response):
            return False, "C005_RESPONSE_SCHEMA_INVALID"

        if (
            response["source_fingerprint"] != fp
            or response["version_trace"]["source_fingerprint"] != fp
        ):
            return False, "SOURCE_FINGERPRINT_MISMATCH"

        if (
            response["version_trace"]["adapter_version"]
            != identity["adapter_version"]
        ):
            return False, "ADAPTER_VERSION_MISMATCH"

        if response["version_trace"]["release_id"] is not None:
            return False, "DRAFT_CANNOT_DECLARE_ACCEPTED_RELEASE"

        outcome = response["outcome"]
        rule = policy["states"][outcome]
        txt = response["answer_text"]

        if (
            rule["answer_text"] == "REQUIRED"
            and not isinstance(txt, str)
        ):
            return False, "REQUIRED_RESPONSE_TEXT_MISSING"

        if rule["answer_text"] == "NULL" and txt is not None:
            return False, "TEXT_FORBIDDEN_FOR_ERROR"

        if (
            rule["claims_and_citations"] == "EMPTY"
            and (response["claims"] or response["citations"])
        ):
            return False, "CITATIONS_OR_CLAIMS_FORBIDDEN_IN_THIS_STATE"

        if (
            rule["claims_and_citations"] == "LINKED_NONEMPTY"
            and (
                not response["claims"]
                or not response["citations"]
            )
        ):
            return False, "ANSWER_REQUIRES_LINKED_CLAIMS_AND_CITATIONS"

        if (
            rule["error"] == "NULL"
            and response["error"] is not None
        ):
            return False, "NON_ERROR_OUTCOME_HAS_ERROR_OBJECT"

        if (
            rule["error"] == "REQUIRED"
            and (
                response["error"] is None
                or response["error"]["code"] not in error_codes
            )
        ):
            return False, "ERROR_OBJECT_CODE_UNKNOWN_OR_MISSING"

        if outcome == "ERROR" and response["confidence"] is not None:
            return False, "ERROR_HAS_CONFIDENCE_MEASUREMENT"

        claims = {
            x["claim_id"]: x
            for x in response["claims"]
        }
        citations = {
            x["citation_id"]: x
            for x in response["citations"]
        }

        if (
            len(claims) != len(response["claims"])
            or len(citations) != len(response["citations"])
        ):
            return False, "DUPLICATE_CLAIM_OR_CITATION_ID"

        if any(
            not claim["citation_ids"]
            for claim in response["claims"]
        ):
            return False, "UNCITED_CLAIM_IN_ANSWER"

        for claim in response["claims"]:
            for citation_id in claim["citation_ids"]:
                if (
                    citation_id not in citations
                    or claim["claim_id"]
                    not in citations[citation_id]["claim_ids"]
                ):
                    return False, "BROKEN_FORWARD_CLAIM_CITATION_LINK"

        for citation in response["citations"]:
            for claim_id in citation["claim_ids"]:
                if (
                    claim_id not in claims
                    or citation["citation_id"]
                    not in claims[claim_id]["citation_ids"]
                ):
                    return False, "BROKEN_REVERSE_CITATION_CLAIM_LINK"

        return True, "DRAFT_STRUCTURAL_POLICY_VALID_ONLY"

    # All fixtures are synthetic and must never be represented as live output.
    def base(outcome="ANSWER", error_code="INTERNAL_ERROR"):
        show_claims = outcome in ("ANSWER", "PARTIAL")
        return {
            "contract_version": "nb23.response.v1",
            "request_contract_version": "nb23.request.v1",
            "request_id": None,
            "source_fingerprint": fp,
            "outcome": outcome,
            "answer_text": (
                None
                if outcome in ("NO_ANSWER", "ERROR")
                else "نمونهٔ مصنوعی قرارداد"
            ),
            "claims": (
                [{
                    "claim_id": "fixture-claim",
                    "text": "متن ساختگی",
                    "citation_ids": ["fixture-citation"],
                }]
                if show_claims else []
            ),
            "citations": (
                [{
                    "citation_id": "fixture-citation",
                    "claim_ids": ["fixture-claim"],
                }]
                if show_claims else []
            ),
            "confidence": None,
            "version_trace": {
                "source_fingerprint": fp,
                "adapter_version": identity["adapter_version"],
                "release_id": None,
            },
            "provenance": {
                "origin": "SYNTHETIC_CONTRACT_FIXTURE",
                "producer_reference": None,
            },
            "error": (
                {
                    "code": error_code,
                    "message": "خطای ساختگی",
                }
                if outcome == "ERROR" else None
            ),
        }

    positives = [
        ("valid_" + state.lower(), base(state))
        for state in expected_states
    ]

    explained = base("NO_ANSWER")
    explained["answer_text"] = (
        "در منابع حاضر پاسخ کافی وجود ندارد (نمونهٔ ساختگی)."
    )
    positives.append((
        "valid_no_answer_explanation",
        explained
    ))

    for error_code in error_codes:
        positives.append((
            "valid_error_code_" + error_code.lower(),
            base("ERROR", error_code)
        ))

    confidence = base("ANSWER")
    confidence["confidence"] = {
        "value": 0.0,
        "meaning": "synthetic measurement",
        "calibration_scope": "synthetic fixture only",
    }
    positives.append((
        "valid_measured_zero_not_missing",
        confidence
    ))

    negatives = []

    def bad(name, state, mutate):
        item = base(state)
        mutate(item)
        negatives.append((name, item))

    bad(
        "answer_missing_citations", "ANSWER",
        lambda r: r["citations"].clear()
    )
    bad(
        "partial_missing_claims", "PARTIAL",
        lambda r: r["claims"].clear()
    )
    bad(
        "answer_uncited_claim", "ANSWER",
        lambda r: r["claims"][0].update(citation_ids=[])
    )
    bad(
        "no_answer_with_claim", "NO_ANSWER",
        lambda r: r["claims"].append({
            "claim_id": "fake",
            "text": "fake",
            "citation_ids": [],
        })
    )
    bad(
        "clarification_without_prompt", "CLARIFICATION",
        lambda r: r.update(answer_text=None)
    )
    bad(
        "safety_refusal_with_citation", "SAFETY_REFUSAL",
        lambda r: r["citations"].append({
            "citation_id": "bad",
            "claim_ids": ["fake"],
        })
    )
    bad(
        "non_error_has_error", "ANSWER",
        lambda r: r.update(error={
            "code": "INTERNAL_ERROR",
            "message": "test",
        })
    )
    bad(
        "error_without_error_object", "ERROR",
        lambda r: r.update(error=None)
    )
    bad(
        "error_with_text", "ERROR",
        lambda r: r.update(answer_text="fake")
    )
    bad(
        "error_with_confidence", "ERROR",
        lambda r: r.update(confidence={
            "value": 0.5,
            "meaning": "test",
            "calibration_scope": "test",
        })
    )
    bad(
        "error_code_unknown", "ERROR",
        lambda r: r["error"].update(
            code="UNMAPPED_PROVIDER_EVENT"
        )
    )
    bad(
        "provider_unknown_misclassified_as_no_answer",
        "NO_ANSWER",
        lambda r: r.update(error={
            "code": "PROVIDER_OUTCOME_UNKNOWN",
            "message": "test",
        })
    )
    bad(
        "wrong_source_fingerprint", "ANSWER",
        lambda r: r.update(
            source_fingerprint="0" * 64
        )
    )
    bad(
        "wrong_adapter", "ANSWER",
        lambda r: r["version_trace"].update(
            adapter_version="unsupported"
        )
    )
    bad(
        "fake_release_id", "ANSWER",
        lambda r: r["version_trace"].update(
            release_id="unaccepted-release"
        )
    )
    bad(
        "broken_claim_citation", "ANSWER",
        lambda r: r["claims"][0].update(
            citation_ids=["missing-citation"]
        )
    )
    bad(
        "broken_citation_backlink", "ANSWER",
        lambda r: r["citations"][0].update(
            claim_ids=["other-claim"]
        )
    )
    bad(
        "duplicate_citation_id", "ANSWER",
        lambda r: r["citations"].append(
            copy.deepcopy(r["citations"][0])
        )
    )
    bad(
        "invalid_consumer_outcome", "ANSWER",
        lambda r: r.update(outcome="ABSTAIN")
    )
    bad(
        "missing_required_version", "ANSWER",
        lambda r: r.pop("contract_version")
    )
    bad(
        "unknown_response_field", "ANSWER",
        lambda r: r.update(http_status=200)
    )
    bad(
        "wrong_provenance_origin", "ANSWER",
        lambda r: r["provenance"].update(
            origin="UNVERIFIED_LIVE"
        )
    )
    bad(
        "unknown_confidence_scope", "ANSWER",
        lambda r: r.update(confidence={
            "value": 0.9,
            "meaning": "unknown",
        })
    )

    results = []

    for name, item, expected in (
        [(n, x, True) for n, x in positives]
        + [(n, x, False) for n, x in negatives]
    ):
        got, reason = validate_draft_response(item)
        ensure(
            got is expected,
            "C007_FIXTURE_UNEXPECTED_RESULT:"
            + name + ":" + reason
        )
        results.append({
            "name": name,
            "expected_valid": expected,
            "observed_valid": got,
            "result": "PASS",
            "decision_reason": reason,
        })

    ensure(
        len(results) == len(positives) + len(negatives),
        "FIXTURE_CENSUS_CHANGED"
    )

    policy_path = (
        root / "schemas/phase2/rag_status_policy.v1.json"
    )
    policy_bytes = canonical(policy)

    if policy_path.exists():
        ensure(
            policy_path.read_bytes() == policy_bytes,
            "STATUS_POLICY_VERSION_CONFLICT"
        )

    report = {
        "schema_version": "NB23_C007_STATUS_ERROR_CONTRACT_REPORT_V1",
        "action": 5,
        "source_fingerprint": fp,
        "source_ref": selection,
        "source_file_sha256": sha(source_raw),
        "parent_receipts_sha256": {
            "C003": h3,
            "C004": h4,
            "C005": h5,
            "C006": h6,
        },
        "source_contract_sha256": {
            "request": c4["schema_file_sha256"],
            "response": c5["response_schema_file_sha256"],
            "citation": c6["citation_schema_sha256"],
        },
        "status_policy_path": (
            "schemas/phase2/rag_status_policy.v1.json"
        ),
        "status_policy_sha256": sha(policy_bytes),
        "consumer_statuses": actual_states,
        "fixture_results": results,
        "positive_fixture_count": len(positives),
        "negative_fixture_count": len(negatives),
        "fixtures_passed": len(results),
        "fixtures_failed": 0,
        "producer_status_enum_verified": False,
        "automatic_producer_to_consumer_status_mapping_verified": False,
        "historical_generator_text_return_verified": True,
        "clarification_runtime_behavior_verified": False,
        "no_answer_runtime_behavior_verified": False,
        "safety_runtime_behavior_verified": False,
        "confidence_runtime_calibration_verified": False,
        "provider_error_transport_tested": False,
        "http_streaming_tested": False,
        "synthetic_fixture_only": True,
        "release_accepted": False,
        "new_api_calls": 0,
        "git_operations": 0,
    }

    report_bytes = canonical(report)
    report_sha = sha(report_bytes)
    destination = work / (
        "NB23.C007.status_error_contract_test."
        + report_sha + ".json"
    )

    immutable(policy_path, policy_bytes)
    immutable(destination, report_bytes)

    ensure(
        decode(policy_path.read_bytes()) == policy
        and decode(destination.read_bytes()) == report
        and sha(destination.read_bytes()) == report_sha,
        "C007_RELOAD_VALIDATION_FAILED"
    )

    return {
        "status": "NB23_C007_DRAFT_STATUS_ERROR_CONTRACT_TESTED",
        "action": 5,
        "source_fingerprint": fp,
        "positive_fixtures": len(positives),
        "negative_fixtures": len(negatives),
        "tests_passed": len(results),
        "status_policy_path": report["status_policy_path"],
        "status_policy_sha256": sha(policy_bytes),
        "producer_enum_mapping": (
            "NOT_VERIFIED_TEXT_ONLY_GENERATOR_INTERFACE"
        ),
        "provider_error_transport": "NOT_TESTED",
        "release_accepted": False,
        "report_path": destination.relative_to(root).as_posix(),
        "report_sha256": report_sha,
        "reload_validation": "PASS",
        "new_api_calls": 0,
        "git_operations": 0,
    }


nb23_c007_result = nb23_c007_status_error_contract()
print(json.dumps(nb23_c007_result, ensure_ascii=False, indent=2))
```
