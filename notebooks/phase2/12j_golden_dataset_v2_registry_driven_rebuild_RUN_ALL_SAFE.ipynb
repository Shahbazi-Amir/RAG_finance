````python
# Code Cell 01

from pathlib import Path
from collections import Counter, defaultdict
from concurrent.futures import ThreadPoolExecutor, as_completed

import ast
import hashlib
import importlib.metadata
import json
import os
import re
import subprocess
import threading
import time

from openai import OpenAI
````

````python
# Code Cell 02

# =========================================================
# Helpers
# =========================================================

def sha256_file(
    path,
):

    digest = hashlib.sha256()

    with path.open(
        "rb",
    ) as file:

        for chunk in iter(
            lambda: file.read(
                1024 * 1024
            ),
            b"",
        ):

            digest.update(
                chunk
            )

    return digest.hexdigest()


def run_git(
    project_root,
    *args,
):

    result = subprocess.run(
        [
            "git",
            *args,
        ],
        cwd=project_root,
        check=True,
        capture_output=True,
        text=True,
    )

    return result.stdout.strip()


def load_json_file(
    path,
):

    with path.open(
        "r",
        encoding="utf-8",
    ) as file:

        return json.load(
            file
        )


def count_jsonl_records(
    path,
):

    count = 0

    with path.open(
        "r",
        encoding="utf-8",
    ) as file:

        for line_number, line in enumerate(
            file,
            start=1,
        ):

            if not line.strip():
                continue

            try:

                json.loads(
                    line
                )

            except json.JSONDecodeError as exc:

                raise AssertionError(
                    "Invalid JSONL record in "
                    f"{path} at line {line_number}"
                ) from exc

            count += 1

    return count


# =========================================================
# Discover Project Root
# =========================================================

current_working_directory = (
    Path.cwd().resolve()
)


project_root_candidates = []


for candidate in [
    current_working_directory,
    *current_working_directory.parents,
]:

    if (
        (
            candidate
            / ".git"
        ).exists()
        and (
            candidate
            / "notebooks"
            / "phase2"
        ).exists()
    ):

        project_root_candidates.append(
            candidate
        )


assert (
    len(
        project_root_candidates
    )
    == 1
), (
    "Could not resolve exactly one project root. "
    f"Candidates: {project_root_candidates}"
)


project_root = (
    project_root_candidates[
        0
    ]
)


git_toplevel = Path(
    run_git(
        project_root,
        "rev-parse",
        "--show-toplevel",
    )
).resolve()


assert (
    git_toplevel
    == project_root
), (
    "Filesystem project root and Git root differ."
)


# =========================================================
# Git read-only state
# =========================================================

git_branch = run_git(
    project_root,
    "rev-parse",
    "--abbrev-ref",
    "HEAD",
)


git_head = run_git(
    project_root,
    "rev-parse",
    "HEAD",
)


git_status_short = run_git(
    project_root,
    "status",
    "--short",
)


# =========================================================
# NB12j identity + historical 12-series inventory
#
# IMPORTANT:
# - Historical upstream notebooks are an explicit frozen set.
# - Any additional 12j copy/backup/repaired file is NOT treated
#   as a historical upstream notebook.
# - This makes Restart Kernel -> Run All safe even if another
#   12j copy exists beside the canonical notebook.
# =========================================================

phase2_notebooks_dir = (
    project_root
    / "notebooks"
    / "phase2"
)


canonical_notebook_path = (
    phase2_notebooks_dir
    / "12j_golden_dataset_v2_registry_driven_rebuild.ipynb"
)


twelve_series_notebooks = sorted(
    path
    for path in phase2_notebooks_dir.glob(
        "12*.ipynb"
    )
    if path.is_file()
)


assert (
    twelve_series_notebooks
), (
    "No 12-series Notebooks found."
)


nb12j_notebook_candidates = sorted(
    path
    for path in twelve_series_notebooks
    if path.name.startswith(
        "12j"
    )
)


assert (
    nb12j_notebook_candidates
), (
    "No NB12j notebook file was found in notebooks/phase2."
)


# Prefer the canonical production filename when it exists.
# Otherwise use the only discovered NB12j candidate.
if canonical_notebook_path.exists():

    notebook_path = (
        canonical_notebook_path
    )

else:

    assert (
        len(
            nb12j_notebook_candidates
        )
        == 1
    ), (
        "Canonical NB12j filename is missing and multiple "
        "NB12j variants exist. Rename the active notebook to:\n"
        f"{canonical_notebook_path}"
    )

    notebook_path = (
        nb12j_notebook_candidates[
            0
        ]
    )


historical_twelve_series_names = [
    "12_golden_dataset.ipynb",
    "12b_golden_dataset_continuation_compact.ipynb",
    "12c_human_review_for_author.ipynb",
    "12d_second_human_review.ipynb",
    "12e_golden_dataset_full_generation.ipynb",
    "12f_author_feedback_integration.ipynb",
    "12g_golden_dataset_quality_rebuild_lab.ipynb",
    "12g_golden_dataset_rebuild_clean.ipynb",
    "12h_pilot_and_expansion_audit.ipynb",
    "12i_claim_registry_and_knowledge_graph.ipynb",
]


historical_twelve_series_notebooks = [
    phase2_notebooks_dir
    / filename
    for filename in (
        historical_twelve_series_names
    )
]


missing_historical_twelve_series_notebooks = [
    path
    for path in (
        historical_twelve_series_notebooks
    )
    if not path.exists()
]


assert not (
    missing_historical_twelve_series_notebooks
), (
    "Missing expected historical 12-series notebooks:\n- "
    + "\n- ".join(
        str(
            path.relative_to(
                project_root
            )
        )
        for path in (
            missing_historical_twelve_series_notebooks
        )
    )
)


assert (
    len(
        historical_twelve_series_notebooks
    )
    == 10
)


historical_twelve_series_relative_paths = {
    str(
        path.relative_to(
            project_root
        )
    )
    for path in (
        historical_twelve_series_notebooks
    )
}


ignored_extra_twelve_series_notebooks = [
    path
    for path in (
        twelve_series_notebooks
    )
    if (
        str(
            path.relative_to(
                project_root
            )
        )
        not in historical_twelve_series_relative_paths
        and path.resolve()
        != notebook_path.resolve()
    )
]


historical_twelve_series_notebook_sha256 = {
    str(
        path.relative_to(
            project_root
        )
    ): sha256_file(
        path
    )
    for path in (
        historical_twelve_series_notebooks
    )
}


# Backward-compatible name used by later cells.
# It intentionally contains ONLY historical upstream notebooks.
twelve_series_notebook_sha256 = dict(
    historical_twelve_series_notebook_sha256
)


# =========================================================
# Final Registry paths from NB12i
# =========================================================

registry_root = (
    project_root
    / "artifacts"
    / "phase2"
    / "knowledge_registry"
)


registry_paths = {
    "contract": (
        registry_root
        / "claim_registry_contract_v1.json"
    ),

    "canonical_claims": (
        registry_root
        / "claims_deduplicated_v1.jsonl"
    ),

    "evidence_links": (
        registry_root
        / "claim_evidence_links_v1.jsonl"
    ),

    "claim_relations": (
        registry_root
        / "claim_relations_v1.jsonl"
    ),

    "knowledge_graph": (
        registry_root
        / "knowledge_graph_v1.json"
    ),

    "registry_report": (
        registry_root
        / "claim_registry_report_v1.json"
    ),

    "claim_id_mapping": (
        registry_root
        / "claim_id_mapping_v1.jsonl"
    ),
}


missing_registry_artifacts = [
    name
    for (
        name,
        path,
    ) in (
        registry_paths.items()
    )
    if not path.exists()
]


assert not missing_registry_artifacts, (
    "Missing final Registry artifacts:\n- "
    + "\n- ".join(
        missing_registry_artifacts
    )
)


# =========================================================
# Verify known frozen Registry hashes
# =========================================================

known_registry_sha256 = {
    "canonical_claims": (
        "8d1080d6519faa8ca5d4614256dbabf04426170b8f914f606eb3088d1f2c08bb"
    ),

    "evidence_links": (
        "1c47a493b2128b1fd489752378a3e5d4dc7e53b7379b03674abe37d9a77cc9ba"
    ),

    "claim_relations": (
        "b505a90036ea9eeec36ddb3e9ebd687f5503c4f8cfccf59f0fb0b684774d9009"
    ),

    "knowledge_graph": (
        "a408f12bac5470de46d700a3c8082e39899473984651b6ca1eeb377e8d80dd1a"
    ),

    "registry_report": (
        "f7951467ca49d1bac1e1c1efee9e93e756f24fbf6db02e500e35a6672b21d645"
    ),

    "claim_id_mapping": (
        "5b5e07134961212984e480e0574722ee2e12cef4aebf21922d7a99a8ab3c9ad2"
    ),
}


registry_sha256 = {
    name: sha256_file(
        path
    )
    for (
        name,
        path,
    ) in (
        registry_paths.items()
    )
}


for (
    artifact_name,
    expected_sha256,
) in (
    known_registry_sha256.items()
):

    actual_sha256 = (
        registry_sha256[
            artifact_name
        ]
    )

    assert (
        actual_sha256
        == expected_sha256
    ), (
        "Frozen Registry SHA mismatch:\n"
        f"Artifact: {artifact_name}\n"
        f"Expected: {expected_sha256}\n"
        f"Actual:   {actual_sha256}"
    )


# =========================================================
# Validate Final Registry Report
# =========================================================

registry_report = load_json_file(
    registry_paths[
        "registry_report"
    ]
)


assert (
    registry_report[
        "registry_status"
    ]
    == "PASS"
)


assert all(
    registry_report[
        "acceptance_checks"
    ].values()
)


registry_counts = (
    registry_report[
        "counts"
    ]
)


assert (
    registry_counts[
        "raw_claims"
    ]
    == 6237
)


assert (
    registry_counts[
        "validated_claims"
    ]
    == 4039
)


assert (
    registry_counts[
        "canonical_claims"
    ]
    == 4006
)


assert (
    registry_counts[
        "validated_registry_evidence_occurrences"
    ]
    == 4125
)


assert (
    registry_counts[
        "claim_relations_total"
    ]
    == 2516
)


assert (
    registry_counts[
        "graph_edge_eligible_claim_relations"
    ]
    == 2415
)


# =========================================================
# Discover legacy Golden / Core / Author Review
# =========================================================

search_roots = [
    project_root
    / "notebooks",

    project_root
    / "artifacts",

    project_root
    / "archive",

    project_root
    / "data",
]


search_roots = [
    path
    for path in search_roots
    if path.exists()
]


def find_unique_filename(
    filename,
):

    matches = []

    for search_root in (
        search_roots
    ):

        matches.extend(
            path.resolve()
            for path in (
                search_root.rglob(
                    filename
                )
            )
            if path.is_file()
        )

    matches = sorted(
        set(
            matches
        )
    )


    assert (
        len(
            matches
        )
        == 1
    ), (
        "Expected exactly one file named "
        f"{filename!r}, found {len(matches)}:\n"
        + "\n".join(
            str(
                path
            )
            for path in matches
        )
    )


    return matches[
        0
    ]


golden_v1_path = (
    find_unique_filename(
        "golden_dataset_v1.jsonl"
    )
)


golden_v1_report_path = (
    find_unique_filename(
        "golden_dataset_v1_final_report.json"
    )
)


core46_path = (
    find_unique_filename(
        "golden_dataset_quote_grounded_final_adjudicated_experiment.jsonl"
    )
)


author_review_path = (
    find_unique_filename(
        "author_review_28_pilots.md"
    )
)


# =========================================================
# Validate legacy JSONL artifacts structurally
# =========================================================

golden_v1_record_count = (
    count_jsonl_records(
        golden_v1_path
    )
)


core46_record_count = (
    count_jsonl_records(
        core46_path
    )
)


golden_v1_sha256 = sha256_file(
    golden_v1_path
)


golden_v1_report_sha256 = sha256_file(
    golden_v1_report_path
)


core46_sha256 = sha256_file(
    core46_path
)


author_review_sha256 = sha256_file(
    author_review_path
)


# =========================================================
# Preflight Gate
# =========================================================

preflight_status = (
    "PASS"
)


next_stage_readiness = (
    "READY_FOR_12_SERIES_CONSOLIDATION_CONTRACT"
)


# =========================================================
# Result
# =========================================================

print(
    "=== NB12j REPOSITORY / ARTIFACT PREFLIGHT ==="
)

print()

print(
    "Project Root:",
    project_root,
)

print(
    "Current Working Directory:",
    current_working_directory,
)

print(
    "Git Branch:",
    git_branch,
)

print(
    "Git HEAD:",
    git_head,
)

print(
    "Git working tree dirty:",
    bool(
        git_status_short
    ),
)


print()

print(
    "Canonical NB12j path:",
    notebook_path.relative_to(
        project_root
    ),
)

print(
    "NB12j variants discovered:",
    len(
        nb12j_notebook_candidates
    ),
)

for path in (
    nb12j_notebook_candidates
):

    print(
        "  -",
        path.relative_to(
            project_root
        ),
    )


print()

print(
    "Historical 12-series Notebooks:",
    len(
        historical_twelve_series_notebooks
    ),
)


for path in (
    historical_twelve_series_notebooks
):

    relative_path = str(
        path.relative_to(
            project_root
        )
    )

    print(
        "  -",
        relative_path,
        historical_twelve_series_notebook_sha256[
            relative_path
        ],
    )


print()

print(
    "Ignored extra 12-series notebook variants:",
    len(
        ignored_extra_twelve_series_notebooks
    ),
)

for path in (
    ignored_extra_twelve_series_notebooks
):

    print(
        "  -",
        path.relative_to(
            project_root
        ),
    )


print()

print(
    "Final Registry status:",
    registry_report[
        "registry_status"
    ],
)

print(
    "Final Registry acceptance checks:",
    sum(
        registry_report[
            "acceptance_checks"
        ].values()
    ),
    "/",
    len(
        registry_report[
            "acceptance_checks"
        ]
    ),
    "PASS",
)


print()

print(
    "Canonical Claims:",
    registry_counts[
        "canonical_claims"
    ],
)

print(
    "Validated Evidence occurrences:",
    registry_counts[
        "validated_registry_evidence_occurrences"
    ],
)

print(
    "Claim Relations:",
    registry_counts[
        "claim_relations_total"
    ],
)

print(
    "Graph-edge eligible Relations:",
    registry_counts[
        "graph_edge_eligible_claim_relations"
    ],
)


print()

print(
    "Golden v1:",
    golden_v1_path.relative_to(
        project_root
    ),
)

print(
    "Golden v1 records:",
    golden_v1_record_count,
)

print(
    "Golden v1 SHA-256:",
    golden_v1_sha256,
)


print()

print(
    "Golden v1 Final Report:",
    golden_v1_report_path.relative_to(
        project_root
    ),
)

print(
    "Golden v1 Final Report SHA-256:",
    golden_v1_report_sha256,
)


print()

print(
    "Core-46:",
    core46_path.relative_to(
        project_root
    ),
)

print(
    "Core-46 records:",
    core46_record_count,
)

print(
    "Core-46 SHA-256:",
    core46_sha256,
)


print()

print(
    "Author Review:",
    author_review_path.relative_to(
        project_root
    ),
)

print(
    "Author Review SHA-256:",
    author_review_sha256,
)


print()

print(
    "Files modified by this Cell:",
    0,
)

print(
    "API calls performed:",
    0,
)


print()

print(
    "Next stage:",
    "12_SERIES_CONSOLIDATION_CONTRACT",
)

print(
    "Readiness:",
    next_stage_readiness,
)


print()

print(
    "CODE CELL 02 GATE:",
    preflight_status,
)
````

````python
# Code Cell 03

assert (
    preflight_status
    == "PASS"
), (
    "Code Cell 02 must PASS before "
    "freezing the NB12j Contract."
)

assert (
    next_stage_readiness
    == "READY_FOR_12_SERIES_CONSOLIDATION_CONTRACT"
)


# =========================================================
# Output namespace
# =========================================================

golden_v2_artifact_root = (
    project_root
    / "artifacts"
    / "phase2"
    / "golden_dataset_v2"
)


golden_v2_artifact_root.mkdir(
    parents=True,
    exist_ok=True,
)


golden_v2_contract_path = (
    golden_v2_artifact_root
    / "golden_dataset_v2_registry_driven_contract_v1.json"
)


# =========================================================
# Frozen publishing helper
#
# Existing file:
#   identical -> REUSED_EXISTING_IDENTICAL
#   different -> FAIL
#
# No destructive overwrite.
# =========================================================

def publish_frozen_text(
    path,
    text,
):

    encoded = text.encode(
        "utf-8"
    )


    if path.exists():

        existing = path.read_bytes()

        assert (
            existing
            == encoded
        ), (
            "Frozen artifact already exists with "
            "different content:\n"
            f"{path}"
        )

        return (
            "REUSED_EXISTING_IDENTICAL"
        )


    path.write_bytes(
        encoded
    )

    return (
        "CREATED_NEW"
    )


# =========================================================
# Source manifests from actual Preflight
# =========================================================

current_notebook_relative_path = (
    "notebooks/phase2/"
    "12j_golden_dataset_v2_registry_driven_rebuild.ipynb"
)


# Freeze hashes only for the explicit historical upstream set.
# Any active/backup/repaired NB12j file is intentionally excluded.
twelve_series_source_manifest = {
    relative_path: {
        "sha256": sha256_value,
        "execution_policy": (
            "DO_NOT_RERUN_FOR_CONSOLIDATION"
        ),
        "mutation_policy": (
            "READ_ONLY"
        ),
    }
    for (
        relative_path,
        sha256_value,
    ) in sorted(
        historical_twelve_series_notebook_sha256.items()
    )
}


assert (
    len(
        twelve_series_source_manifest
    )
    == 10
), (
    "Historical 12-series source manifest must contain "
    "exactly the 10 validated upstream notebooks."
)


registry_source_manifest = {
    name: {
        "path": str(
            path.relative_to(
                project_root
            )
        ),
        "sha256": (
            registry_sha256[
                name
            ]
        ),
    }
    for (
        name,
        path,
    ) in sorted(
        registry_paths.items()
    )
}


legacy_golden_source_manifest = {
    "golden_v1": {
        "path": str(
            golden_v1_path.relative_to(
                project_root
            )
        ),
        "sha256": (
            golden_v1_sha256
        ),
        "records": (
            golden_v1_record_count
        ),
        "role": (
            "HISTORICAL_BASELINE"
        ),
        "mutation_policy": (
            "IMMUTABLE"
        ),
    },

    "golden_v1_final_report": {
        "path": str(
            golden_v1_report_path.relative_to(
                project_root
            )
        ),
        "sha256": (
            golden_v1_report_sha256
        ),
        "role": (
            "HISTORICAL_BASELINE_REPORT"
        ),
        "mutation_policy": (
            "IMMUTABLE"
        ),
    },

    "core46": {
        "path": str(
            core46_path.relative_to(
                project_root
            )
        ),
        "sha256": (
            core46_sha256
        ),
        "records": (
            core46_record_count
        ),
        "role": (
            "QUALITY_AND_REGRESSION_ANCHOR"
        ),
        "mutation_policy": (
            "IMMUTABLE"
        ),
    },

    "author_review": {
        "path": str(
            author_review_path.relative_to(
                project_root
            )
        ),
        "sha256": (
            author_review_sha256
        ),
        "role": (
            "AUTHOR_REVIEW_SOURCE"
        ),
        "review_status": (
            "TO_BE_DISCOVERED_FROM_SOURCE"
        ),
        "mutation_policy": (
            "IMMUTABLE"
        ),
    },
}


# =========================================================
# Planned artifact family
#
# These are NEW NB12j-owned outputs.
# Previous 12-series artifacts remain untouched.
# =========================================================

golden_v2_artifact_family = {
    "contract": (
        "golden_dataset_v2_registry_driven_contract_v1.json"
    ),

    "twelve_series_manifest": (
        "twelve_series_consolidated_manifest_v1.json"
    ),

    "claim_eligibility_manifest": (
        "golden_v2_claim_eligibility_manifest_v1.jsonl"
    ),

    "seed_pool": (
        "golden_v2_seed_pool_v1.jsonl"
    ),

    "dataset_candidate": (
        "golden_dataset_v2_registry_driven_candidate_v1.jsonl"
    ),

    "candidate_report": (
        "golden_dataset_v2_registry_driven_candidate_report_v1.json"
    ),

    "final_acceptance_report": (
        "golden_dataset_v2_final_acceptance_report_v1.json"
    ),
}


# =========================================================
# Contract
# =========================================================

golden_v2_contract = {
    "contract_version": (
        "golden_dataset_v2_registry_driven_contract_v1"
    ),

    "notebook": (
        "notebooks/phase2/"
        "12j_golden_dataset_v2_registry_driven_rebuild.ipynb"
    ),

    "purpose": (
        "Consolidate the validated outputs of the "
        "12-series workflow and build a broad, "
        "registry-driven, evidence-grounded Golden "
        "Dataset v2 without modifying frozen upstream "
        "artifacts."
    ),

    "pipeline_position": [
        "SOURCE_CORPUS",
        "CLAIM_REGISTRY_AND_KNOWLEDGE_GRAPH",
        "GOLDEN_DATASET_V2",
        "BENCHMARK",
    ],

    "source_state": {
        "git_branch": (
            git_branch
        ),

        "git_head_at_preflight": (
            git_head
        ),

        "historical_twelve_series_notebook_count": (
            len(
                historical_twelve_series_notebooks
            )
        ),

        "nb12j_variant_count_at_preflight": (
            len(
                nb12j_notebook_candidates
            )
        ),

        "registry_status": (
            registry_report[
                "registry_status"
            ]
        ),

        "registry_acceptance_checks_passed": (
            sum(
                registry_report[
                    "acceptance_checks"
                ].values()
            )
        ),

        "registry_acceptance_checks_total": (
            len(
                registry_report[
                    "acceptance_checks"
                ]
            )
        ),

        "canonical_claims": (
            registry_counts[
                "canonical_claims"
            ]
        ),

        "validated_evidence_occurrences": (
            registry_counts[
                "validated_registry_evidence_occurrences"
            ]
        ),

        "claim_relations": (
            registry_counts[
                "claim_relations_total"
            ]
        ),
    },

    "source_manifests": {
        "twelve_series_notebooks": (
            twelve_series_source_manifest
        ),

        "claim_registry": (
            registry_source_manifest
        ),

        "legacy_golden_and_review": (
            legacy_golden_source_manifest
        ),
    },

    "consolidation_policy": {
        "rerun_old_12_series_notebooks": False,

        "copy_old_notebook_code_into_12j": False,

        "modify_old_12_series_notebooks": False,

        "modify_old_artifacts": False,

        "status_must_be_evidence_based": True,

        "unknown_status_when_not_provable": True,

        "preserve_source_hashes": True,

        "consolidation_unit": (
            "NOTEBOOK_AND_ARTIFACT_PROVENANCE"
        ),
    },

    "registry_consumption_policy": {
        "primary_seed_population_artifact": (
            "claims_deduplicated_v1.jsonl"
        ),

        "canonical_claim_population": (
            registry_counts[
                "canonical_claims"
            ]
        ),

        "raw_claims_are_direct_seed_population": False,

        "claim_id_required": True,

        "evidence_required": True,

        "source_provenance_required": True,

        "automatic_semantic_merge_allowed": False,

        "automatic_subsumption_merge_allowed": False,

        "original_source_remains_final_evidence": True,

        "retrieval_chunks_define_golden_truth": False,
    },

    "golden_seed_policy": {
        "seed_unit": (
            "INDEPENDENT_EVIDENCE_GROUNDED_CANONICAL_CLAIM"
        ),

        "paraphrase_counts_as_new_seed": False,

        "question_variant_counts_as_new_seed": False,

        "seed_requires_canonical_claim_id": True,

        "seed_requires_validated_evidence": True,

        "seed_requires_traceable_source": True,

        "seed_count_target": (
            "TO_BE_DERIVED_FROM_ACTUAL_ELIGIBILITY_AND_COVERAGE"
        ),

        "do_not_invent_fixed_seed_count": True,
    },

    "legacy_golden_policy": {
        "golden_v1_role": (
            "HISTORICAL_BASELINE"
        ),

        "golden_v1_mutable": False,

        "core46_role": (
            "QUALITY_AND_REGRESSION_ANCHOR"
        ),

        "core46_mutable": False,

        "core46_is_full_knowledge_universe": False,
    },

    "author_review_policy": {
        "status_source": (
            "author_review_28_pilots.md"
        ),

        "status_must_be_discovered": True,

        "assume_feedback_complete": False,

        "engineering_candidate_allowed_while_pending": True,

        "author_reviewed_final_freeze_requires_real_feedback": True,

        "fabricate_author_feedback": False,
    },

    "golden_generation_policy": {
        "evidence_first": True,

        "answer_must_be_grounded": True,

        "question_must_be_supported_by_claim_and_evidence": True,

        "preserve_claim_id": True,

        "preserve_evidence_id": True,

        "preserve_source_lineage": True,

        "control_seed_redundancy": True,

        "control_question_redundancy": True,

        "measure_registry_coverage": True,

        "measure_source_coverage": True,

        "measure_claim_type_coverage": True,

        "measure_authority_coverage": True,

        "negative_or_challenge_cases_must_not_break_grounding": True,
    },

    "artifact_namespace": str(
        golden_v2_artifact_root.relative_to(
            project_root
        )
    ),

    "artifact_family": (
        golden_v2_artifact_family
    ),

    "write_policy": {
        "versioned_outputs": True,

        "destructive_overwrite": False,

        "reuse_existing_only_if_byte_identical": True,

        "git_write_without_explicit_permission": False,
    },

    "final_freeze_policy": {
        "engineering_acceptance_required": True,

        "restart_kernel_run_all_required": True,

        "final_notebook_health_required": True,

        "author_reviewed_label_requires_completed_author_feedback": True,
    },

    "current_state": (
        "CONTRACT_FROZEN_PENDING_12_SERIES_EVIDENCE_DISCOVERY"
    ),
}


# =========================================================
# Deterministic serialization / rerun-safe freeze
# =========================================================

proposed_golden_v2_contract_text = (
    json.dumps(
        golden_v2_contract,
        ensure_ascii=False,
        indent=2,
        sort_keys=True,
    )
    + "\n"
)


def validate_existing_golden_v2_contract(
    existing_contract,
):

    assert (
        existing_contract.get(
            "contract_version"
        )
        == "golden_dataset_v2_registry_driven_contract_v1"
    )

    assert (
        existing_contract.get(
            "notebook"
        )
        == current_notebook_relative_path
    )

    source_state = existing_contract.get(
        "source_state",
        {},
    )

    assert (
        source_state.get(
            "canonical_claims"
        )
        == 4006
    )

    assert (
        source_state.get(
            "registry_status"
        )
        == "PASS"
    )

    assert (
        existing_contract.get(
            "artifact_family"
        )
        == golden_v2_artifact_family
    )

    existing_source_manifests = existing_contract.get(
        "source_manifests",
        {},
    )

    assert (
        existing_source_manifests.get(
            "claim_registry"
        )
        == registry_source_manifest
    ), (
        "Existing frozen contract no longer matches the "
        "frozen Claim Registry inputs."
    )

    assert (
        existing_source_manifests.get(
            "legacy_golden_and_review"
        )
        == legacy_golden_source_manifest
    ), (
        "Existing frozen contract no longer matches the "
        "legacy Golden / review inputs."
    )

    existing_twelve_series = existing_source_manifests.get(
        "twelve_series_notebooks",
        {},
    )

    # Historical upstream notebooks must still match exactly.
    # An older valid v1 contract may also contain a hash for NB12j
    # itself; that self-referential entry is intentionally ignored
    # during rerun validation because NB12j is the active notebook.
    for (
        relative_path,
        expected_entry,
    ) in twelve_series_source_manifest.items():

        assert (
            relative_path
            in existing_twelve_series
        ), (
            "Existing contract is missing historical notebook: "
            f"{relative_path}"
        )

        assert (
            existing_twelve_series[
                relative_path
            ][
                "sha256"
            ]
            == expected_entry[
                "sha256"
            ]
        ), (
            "Historical notebook hash changed after contract freeze: "
            f"{relative_path}"
        )

    assert (
        existing_contract[
            "golden_seed_policy"
        ][
            "paraphrase_counts_as_new_seed"
        ]
        is False
    )

    assert (
        existing_contract[
            "author_review_policy"
        ][
            "assume_feedback_complete"
        ]
        is False
    )


if golden_v2_contract_path.exists():

    disk_golden_v2_contract = load_json_file(
        golden_v2_contract_path
    )

    validate_existing_golden_v2_contract(
        disk_golden_v2_contract
    )

    golden_v2_contract = (
        disk_golden_v2_contract
    )

    golden_v2_contract_text = (
        golden_v2_contract_path.read_text(
            encoding="utf-8"
        )
    )

    golden_v2_contract_sha256 = sha256_file(
        golden_v2_contract_path
    )

    golden_v2_contract_write_status = (
        "REUSED_EXISTING_VALIDATED"
    )

else:

    golden_v2_contract_text = (
        proposed_golden_v2_contract_text
    )

    golden_v2_contract_sha256 = (
        hashlib.sha256(
            golden_v2_contract_text.encode(
                "utf-8"
            )
        ).hexdigest()
    )

    golden_v2_contract_write_status = (
        publish_frozen_text(
            golden_v2_contract_path,
            golden_v2_contract_text,
        )
    )

    disk_golden_v2_contract = load_json_file(
        golden_v2_contract_path
    )

    assert (
        disk_golden_v2_contract
        == golden_v2_contract
    )

    assert (
        sha256_file(
            golden_v2_contract_path
        )
        == golden_v2_contract_sha256
    )


# =========================================================
# Independent disk revalidation
# =========================================================

assert (
    sha256_file(
        golden_v2_contract_path
    )
    == golden_v2_contract_sha256
)

assert (
    golden_v2_contract[
        "source_state"
    ][
        "canonical_claims"
    ]
    == 4006
)

assert (
    golden_v2_contract[
        "source_state"
    ][
        "registry_status"
    ]
    == "PASS"
)

assert (
    golden_v2_contract[
        "golden_seed_policy"
    ][
        "paraphrase_counts_as_new_seed"
    ]
    is False
)

assert (
    golden_v2_contract[
        "author_review_policy"
    ][
        "assume_feedback_complete"
    ]
    is False
)


# =========================================================
# Gate
# =========================================================

consolidation_contract_status = (
    "PASS"
)


twelve_series_evidence_discovery_readiness = (
    "READY_FOR_12_SERIES_EVIDENCE_DISCOVERY"
)


# =========================================================
# Result
# =========================================================

print(
    "=== NB12j CONSOLIDATION / GOLDEN V2 CONTRACT ==="
)

print()

print(
    "Historical 12-series source Notebooks hashed:",
    len(
        twelve_series_source_manifest
    ),
)

print(
    "Historical 12-series upstream Notebooks:",
    len(
        historical_twelve_series_notebooks
    ),
)

print(
    "NB12j variants present:",
    len(
        nb12j_notebook_candidates
    ),
)

print(
    "Canonical Registry Claims:",
    registry_counts[
        "canonical_claims"
    ],
)

print(
    "Golden v1 baseline records:",
    golden_v1_record_count,
)

print(
    "Core-46 anchor records:",
    core46_record_count,
)


print()

print(
    "Output namespace:",
    golden_v2_artifact_root.relative_to(
        project_root
    ),
)

print(
    "Planned NB12j artifact family:",
    len(
        golden_v2_artifact_family
    ),
)


for (
    role,
    filename,
) in sorted(
    golden_v2_artifact_family.items()
):

    print(
        "  -",
        role,
        "->",
        filename,
    )


print()

print(
    "Old 12-series notebooks mutable:",
    False,
)

print(
    "Old Golden artifacts mutable:",
    False,
)

print(
    "Raw Claims used directly as Seeds:",
    False,
)

print(
    "Paraphrase counts as new Seed:",
    False,
)

print(
    "Author Feedback assumed complete:",
    False,
)

print(
    "Git writes performed:",
    0,
)

print(
    "API calls performed:",
    0,
)


print()

print(
    "Contract:",
    golden_v2_contract_path.relative_to(
        project_root
    ),
)

print(
    "Contract SHA-256:",
    golden_v2_contract_sha256,
)

print(
    "Write status:",
    golden_v2_contract_write_status,
)


print()

print(
    "Next stage:",
    "12_SERIES_EVIDENCE_DISCOVERY",
)

print(
    "Readiness:",
    twelve_series_evidence_discovery_readiness,
)


print()

print(
    "CODE CELL 03 GATE:",
    consolidation_contract_status,
)
````

````python
# Code Cell 04

assert (
    consolidation_contract_status
    == "PASS"
), (
    "Code Cell 03 must PASS before "
    "12-series Evidence Discovery."
)

assert (
    twelve_series_evidence_discovery_readiness
    == "READY_FOR_12_SERIES_EVIDENCE_DISCOVERY"
)


# =========================================================
# Helpers for .ipynb inspection
# =========================================================

def cell_source_text(
    cell,
):

    source = cell.get(
        "source",
        "",
    )

    if isinstance(
        source,
        list,
    ):

        source = "".join(
            source
        )

    return source


def output_text(
    output,
):

    output_type = output.get(
        "output_type"
    )


    if (
        output_type
        == "stream"
    ):

        text = output.get(
            "text",
            "",
        )

        if isinstance(
            text,
            list,
        ):

            text = "".join(
                text
            )

        return text


    if (
        output_type
        in {
            "execute_result",
            "display_data",
        }
    ):

        data = output.get(
            "data",
            {},
        )

        text = data.get(
            "text/plain",
            "",
        )

        if isinstance(
            text,
            list,
        ):

            text = "".join(
                text
            )

        return text


    if (
        output_type
        == "error"
    ):

        return (
            "ERROR "
            + str(
                output.get(
                    "ename",
                    "",
                )
            )
            + ": "
            + str(
                output.get(
                    "evalue",
                    "",
                )
            )
        )


    return ""


def first_markdown_heading(
    notebook_json,
):

    for cell in notebook_json.get(
        "cells",
        []
    ):

        if (
            cell.get(
                "cell_type"
            )
            != "markdown"
        ):

            continue


        source = cell_source_text(
            cell
        )


        # Markdown heading
        match = re.search(
            r"(?m)^\s*#{1,3}\s+(.+?)\s*$",
            source,
        )

        if match:

            return (
                match.group(
                    1
                ).strip()
            )


        # HTML heading
        match = re.search(
            r"<h[1-3][^>]*>(.*?)</h[1-3]>",
            source,
            flags=re.IGNORECASE
            | re.DOTALL,
        )

        if match:

            heading = re.sub(
                r"<[^>]+>",
                "",
                match.group(
                    1
                ),
            )

            return (
                heading.strip()
            )


    return None


def extract_signal_lines(
    text,
):

    signal_patterns = [
        r"\bGATE\b",
        r"\bPASS\b",
        r"\bFAIL\b",
        r"FINAL .*STATUS",
        r"FINAL .*HEALTH",
        r"NOTEBOOK HEALTH",
        r"NOTEBOOK OBJECTIVE",
        r"GOAL_ACHIEVED",
        r"READY_FOR_",
        r"BLOCKED",
        r"PENDING",
        r"FROZEN",
    ]


    combined_pattern = re.compile(
        "|".join(
            signal_patterns
        ),
        flags=re.IGNORECASE,
    )


    signals = []


    for raw_line in text.splitlines():

        line = raw_line.strip()


        if (
            line
            and combined_pattern.search(
                line
            )
        ):

            signals.append(
                line
            )


    # Preserve order, remove duplicates.
    deduplicated = []

    seen = set()


    for line in signals:

        if line in seen:
            continue

        seen.add(
            line
        )

        deduplicated.append(
            line
        )


    return deduplicated


# =========================================================
# Inspect PRIOR 12-series Notebooks
# Current NB12j is excluded from historical discovery.
# =========================================================

# Use the explicit historical upstream set from Code Cell 02.
# Do not derive history by subtracting one active filename from
# a glob, because local backup/repaired NB12j copies may exist.
prior_twelve_series_notebooks = list(
    historical_twelve_series_notebooks
)


assert (
    len(
        prior_twelve_series_notebooks
    )
    == 10
), (
    "Expected the validated historical 12-series set "
    "to contain exactly 10 notebooks."
)


twelve_series_evidence = []


for path in (
    prior_twelve_series_notebooks
):

    notebook_json = (
        load_json_file(
            path
        )
    )


    cells = notebook_json.get(
        "cells",
        []
    )


    code_cells = [
        cell
        for cell in cells
        if (
            cell.get(
                "cell_type"
            )
            == "code"
        )
    ]


    markdown_cells = [
        cell
        for cell in cells
        if (
            cell.get(
                "cell_type"
            )
            == "markdown"
        )
    ]


    executed_code_cells = [
        cell
        for cell in code_cells
        if (
            cell.get(
                "execution_count"
            )
            is not None
        )
    ]


    unexecuted_code_cells = [
        cell
        for cell in code_cells
        if (
            cell.get(
                "execution_count"
            )
            is None
        )
    ]


    error_outputs = []


    stored_output_fragments = []


    for (
        cell_index,
        cell,
    ) in enumerate(
        code_cells,
        start=1,
    ):

        for output in cell.get(
            "outputs",
            []
        ):

            text = output_text(
                output
            )


            if text:

                stored_output_fragments.append(
                    text
                )


            if (
                output.get(
                    "output_type"
                )
                == "error"
            ):

                error_outputs.append({
                    "code_cell_position": (
                        cell_index
                    ),

                    "ename": output.get(
                        "ename"
                    ),

                    "evalue": output.get(
                        "evalue"
                    ),
                })


    stored_outputs_text = "\n".join(
        stored_output_fragments
    )


    signal_lines = extract_signal_lines(
        stored_outputs_text
    )


    execution_counts = [
        cell.get(
            "execution_count"
        )
        for cell in (
            executed_code_cells
        )
        if isinstance(
            cell.get(
                "execution_count"
            ),
            int,
        )
    ]


    last_execution_count = (
        max(
            execution_counts
        )
        if execution_counts
        else None
    )


    twelve_series_evidence.append({
        "path": str(
            path.relative_to(
                project_root
            )
        ),

        "sha256": (
            twelve_series_notebook_sha256[
                str(
                    path.relative_to(
                        project_root
                    )
                )
            ]
        ),

        "first_heading": (
            first_markdown_heading(
                notebook_json
            )
        ),

        "markdown_cells": (
            len(
                markdown_cells
            )
        ),

        "code_cells": (
            len(
                code_cells
            )
        ),

        "executed_code_cells": (
            len(
                executed_code_cells
            )
        ),

        "unexecuted_code_cells": (
            len(
                unexecuted_code_cells
            )
        ),

        "stored_error_outputs": (
            len(
                error_outputs
            )
        ),

        "error_evidence": (
            error_outputs
        ),

        "last_execution_count": (
            last_execution_count
        ),

        "signal_lines": (
            signal_lines
        ),
    })


# =========================================================
# Author Review evidence discovery
# =========================================================

author_review_text = author_review_path.read_text(
    encoding="utf-8"
)


author_review_lines = (
    author_review_text.splitlines()
)


author_review_marker_lines = [
    line.strip()
    for line in author_review_lines
    if (
        "نظر مؤلف"
        in line
    )
]


review_case_markers = [
    line.strip()
    for line in author_review_lines
    if re.search(
        r"(?i)\breview\s*case\b",
        line,
    )
]


# Conservative discovery:
# collect text occurring on the SAME line after "نظر مؤلف".
# We do not infer completion from surrounding structure yet.
author_review_inline_values = []


for line in (
    author_review_marker_lines
):

    marker_position = line.find(
        "نظر مؤلف"
    )


    suffix = line[
        marker_position
        + len(
            "نظر مؤلف"
        ):
    ].strip()


    suffix = suffix.lstrip(
        ":：-*# "
    ).strip()


    if suffix:

        author_review_inline_values.append(
            suffix
        )


# =========================================================
# Golden v1 report structural discovery
# =========================================================

golden_v1_report = load_json_file(
    golden_v1_report_path
)


golden_v1_report_keys = sorted(
    golden_v1_report.keys()
)


# =========================================================
# Golden v1 first-record schema
# =========================================================

def first_jsonl_record(
    path,
):

    with path.open(
        "r",
        encoding="utf-8",
    ) as file:

        for line in file:

            if line.strip():

                return json.loads(
                    line
                )


    raise AssertionError(
        f"No JSONL records found: {path}"
    )


golden_v1_first_record = (
    first_jsonl_record(
        golden_v1_path
    )
)


core46_first_record = (
    first_jsonl_record(
        core46_path
    )
)


golden_v1_record_keys = sorted(
    golden_v1_first_record.keys()
)


core46_record_keys = sorted(
    core46_first_record.keys()
)


# =========================================================
# No writes in this Step
# =========================================================

twelve_series_evidence_discovery_status = (
    "PASS"
)


twelve_series_manifest_readiness = (
    "READY_FOR_12_SERIES_CONSOLIDATED_MANIFEST"
)


# =========================================================
# Result
# =========================================================

print(
    "=== NB12j 12-SERIES EVIDENCE DISCOVERY ==="
)

print()

print(
    "Prior 12-series Notebooks inspected:",
    len(
        twelve_series_evidence
    ),
)


for evidence in (
    twelve_series_evidence
):

    print()
    print(
        evidence[
            "path"
        ]
    )

    print(
        "  Heading:",
        evidence[
            "first_heading"
        ],
    )

    print(
        "  Markdown / Code:",
        evidence[
            "markdown_cells"
        ],
        "/",
        evidence[
            "code_cells"
        ],
    )

    print(
        "  Executed / Unexecuted Code:",
        evidence[
            "executed_code_cells"
        ],
        "/",
        evidence[
            "unexecuted_code_cells"
        ],
    )

    print(
        "  Stored Error outputs:",
        evidence[
            "stored_error_outputs"
        ],
    )

    print(
        "  Last execution count:",
        evidence[
            "last_execution_count"
        ],
    )

    print(
        "  Explicit signal lines:",
        len(
            evidence[
                "signal_lines"
            ]
        ),
    )


    for line in (
        evidence[
            "signal_lines"
        ][
            -12:
        ]
    ):

        print(
            "    ",
            line,
        )


print()
print(
    "=== AUTHOR REVIEW DISCOVERY ==="
)

print(
    "Review Case markers:",
    len(
        review_case_markers
    ),
)

print(
    "'نظر مؤلف' markers:",
    len(
        author_review_marker_lines
    ),
)

print(
    "Non-empty inline values after 'نظر مؤلف':",
    len(
        author_review_inline_values
    ),
)


if author_review_inline_values:

    print(
        "Inline value samples:"
    )

    for value in (
        author_review_inline_values[
            :10
        ]
    ):

        print(
            "  -",
            value,
        )


print()
print(
    "=== LEGACY GOLDEN STRUCTURE ==="
)

print(
    "Golden v1 records:",
    golden_v1_record_count,
)

print(
    "Golden v1 first-record keys:",
    golden_v1_record_keys,
)

print(
    "Golden v1 report top-level keys:",
    golden_v1_report_keys,
)


print()

print(
    "Core-46 records:",
    core46_record_count,
)

print(
    "Core-46 first-record keys:",
    core46_record_keys,
)


print()

print(
    "Files modified by this Cell:",
    0,
)

print(
    "API calls performed:",
    0,
)


print()

print(
    "Next stage:",
    "12_SERIES_CONSOLIDATED_MANIFEST",
)

print(
    "Readiness:",
    twelve_series_manifest_readiness,
)


print()

print(
    "CODE CELL 04 GATE:",
    twelve_series_evidence_discovery_status,
)
````

````python
# Code Cell 06

assert (
    twelve_series_manifest_status
    == "PASS"
), (
    "Code Cell 05 must PASS before "
    "Registry Claim Eligibility Profile."
)

assert (
    claim_eligibility_profile_readiness
    == "READY_FOR_REGISTRY_CLAIM_ELIGIBILITY_PROFILE"
)


# =========================================================
# Helpers
# =========================================================

def load_jsonl_records(
    path,
):

    records = []

    with path.open(
        "r",
        encoding="utf-8",
    ) as file:

        for line_number, line in enumerate(
            file,
            start=1,
        ):

            if not line.strip():
                continue

            try:

                record = json.loads(
                    line
                )

            except json.JSONDecodeError as exc:

                raise AssertionError(
                    "Invalid JSONL record:\n"
                    f"{path}\n"
                    f"Line: {line_number}"
                ) from exc

            records.append(
                record
            )

    return records


def numeric_profile(
    values,
):

    assert values

    ordered = sorted(
        values
    )

    count = len(
        ordered
    )


    def percentile(
        fraction,
    ):

        if count == 1:
            return ordered[0]

        position = (
            count - 1
        ) * fraction

        lower_index = int(
            position
        )

        upper_index = min(
            lower_index + 1,
            count - 1,
        )

        weight = (
            position
            - lower_index
        )

        return (
            ordered[
                lower_index
            ]
            * (
                1
                - weight
            )
            + ordered[
                upper_index
            ]
            * weight
        )


    return {
        "count": (
            count
        ),

        "min": (
            ordered[0]
        ),

        "p50": (
            percentile(
                0.50
            )
        ),

        "p90": (
            percentile(
                0.90
            )
        ),

        "p95": (
            percentile(
                0.95
            )
        ),

        "p99": (
            percentile(
                0.99
            )
        ),

        "max": (
            ordered[-1]
        ),

        "mean": (
            sum(
                ordered
            )
            / count
        ),
    }


def sorted_counter_dict(
    counter,
):

    return dict(
        sorted(
            counter.items(),
            key=lambda item: (
                str(
                    item[0]
                )
            ),
        )
    )


def is_nonempty_value(
    value,
):

    if value is None:
        return False

    if isinstance(
        value,
        str,
    ):

        return bool(
            value.strip()
        )

    if isinstance(
        value,
        (
            list,
            dict,
            tuple,
            set,
        ),
    ):

        return bool(
            value
        )

    return True


# =========================================================
# Reload frozen Registry artifacts independently
# =========================================================

profile_claims = load_jsonl_records(
    registry_paths[
        "canonical_claims"
    ]
)

profile_all_evidence = load_jsonl_records(
    registry_paths[
        "evidence_links"
    ]
)

profile_relations = load_jsonl_records(
    registry_paths[
        "claim_relations"
    ]
)


assert (
    len(
        profile_claims
    )
    == registry_counts[
        "canonical_claims"
    ]
)


assert (
    len(
        profile_relations
    )
    == registry_counts[
        "claim_relations_total"
    ]
)


# =========================================================
# Claim identity validation
# =========================================================

profile_claim_by_id = {}


for claim in (
    profile_claims
):

    claim_id = claim.get(
        "claim_id"
    )

    assert (
        is_nonempty_value(
            claim_id
        )
    )

    assert (
        claim_id
        not in profile_claim_by_id
    ), (
        "Duplicate canonical claim_id: "
        f"{claim_id}"
    )

    profile_claim_by_id[
        claim_id
    ] = claim


assert (
    len(
        profile_claim_by_id
    )
    == 4006
)


# =========================================================
# Evidence identity validation
# =========================================================

profile_evidence_by_id = {}


for evidence in (
    profile_all_evidence
):

    evidence_id = evidence.get(
        "evidence_id"
    )

    assert (
        is_nonempty_value(
            evidence_id
        )
    )

    assert (
        evidence_id
        not in profile_evidence_by_id
    ), (
        "Duplicate evidence_id: "
        f"{evidence_id}"
    )

    profile_evidence_by_id[
        evidence_id
    ] = evidence


registry_evidence_ids = {
    evidence_id
    for claim in (
        profile_claims
    )
    for evidence_id in (
        claim[
            "evidence_ids"
        ]
    )
}


assert (
    len(
        registry_evidence_ids
    )
    == registry_counts[
        "validated_registry_evidence_occurrences"
    ]
)


missing_registry_evidence_ids = sorted(
    registry_evidence_ids
    - set(
        profile_evidence_by_id
    )
)


assert not missing_registry_evidence_ids, (
    "Canonical Registry references missing "
    "Evidence IDs."
)


profile_registry_evidence = [
    profile_evidence_by_id[
        evidence_id
    ]
    for evidence_id in sorted(
        registry_evidence_ids
    )
]


assert (
    len(
        profile_registry_evidence
    )
    == 4125
)


# =========================================================
# Structural eligibility
#
# IMPORTANT:
# This is NOT final Golden Seed eligibility.
# It checks only whether a Canonical Claim is structurally
# ready to ENTER Golden suitability review.
# =========================================================

structural_failures_by_claim = {}


for claim in (
    profile_claims
):

    claim_id = claim[
        "claim_id"
    ]

    failures = []


    if not is_nonempty_value(
        claim.get(
            "claim_text_fa"
        )
    ):

        failures.append(
            "MISSING_CLAIM_TEXT"
        )


    if (
        claim.get(
            "registry_validation_status"
        )
        != "VALIDATED"
    ):

        failures.append(
            "NOT_VALIDATED"
        )


    evidence_ids = claim.get(
        "evidence_ids",
        [],
    )


    if not evidence_ids:

        failures.append(
            "NO_EVIDENCE_IDS"
        )


    missing_evidence_for_claim = [
        evidence_id
        for evidence_id in (
            evidence_ids
        )
        if (
            evidence_id
            not in profile_evidence_by_id
        )
    ]


    if missing_evidence_for_claim:

        failures.append(
            "REFERENCES_MISSING_EVIDENCE"
        )


    if not claim.get(
        "source_ids"
    ):

        failures.append(
            "NO_SOURCE_IDS"
        )


    if not claim.get(
        "source_document_ids"
    ):

        failures.append(
            "NO_SOURCE_DOCUMENT_IDS"
        )


    if not claim.get(
        "source_types"
    ):

        failures.append(
            "NO_SOURCE_TYPES"
        )


    if not claim.get(
        "authority_classes"
    ):

        failures.append(
            "NO_AUTHORITY_CLASSES"
        )


    if not claim.get(
        "claim_discovery_roles"
    ):

        failures.append(
            "NO_DISCOVERY_ROLES"
        )


    if not claim.get(
        "claim_types"
    ):

        failures.append(
            "NO_CLAIM_TYPES"
        )


    if failures:

        structural_failures_by_claim[
            claim_id
        ] = failures


structurally_eligible_claim_ids = sorted(
    set(
        profile_claim_by_id
    )
    - set(
        structural_failures_by_claim
    )
)


structurally_ineligible_claim_ids = sorted(
    structural_failures_by_claim
)


# =========================================================
# Claim Type profile
# =========================================================

claim_type_membership_counts = Counter()

canonical_singular_claim_type_counts = Counter()


for claim in (
    profile_claims
):

    canonical_singular_claim_type_counts[
        str(
            claim.get(
                "claim_type"
            )
        )
    ] += 1


    for claim_type in (
        claim.get(
            "claim_types",
            []
        )
    ):

        claim_type_membership_counts[
            claim_type
        ] += 1


multi_claim_type_claims = [
    claim[
        "claim_id"
    ]
    for claim in (
        profile_claims
    )
    if (
        len(
            claim.get(
                "claim_types",
                []
            )
        )
        > 1
    )
]


# =========================================================
# Authority / discovery role / source type profiles
# =========================================================

authority_membership_counts = Counter()
discovery_role_membership_counts = Counter()
source_type_membership_counts = Counter()


multi_authority_claims = []
multi_discovery_role_claims = []
multi_source_type_claims = []


for claim in (
    profile_claims
):

    authorities = claim.get(
        "authority_classes",
        []
    )

    discovery_roles = claim.get(
        "claim_discovery_roles",
        []
    )

    source_types = claim.get(
        "source_types",
        []
    )


    for authority in authorities:

        authority_membership_counts[
            authority
        ] += 1


    for role in discovery_roles:

        discovery_role_membership_counts[
            role
        ] += 1


    for source_type in source_types:

        source_type_membership_counts[
            source_type
        ] += 1


    if len(
        authorities
    ) > 1:

        multi_authority_claims.append(
            claim[
                "claim_id"
            ]
        )


    if len(
        discovery_roles
    ) > 1:

        multi_discovery_role_claims.append(
            claim[
                "claim_id"
            ]
        )


    if len(
        source_types
    ) > 1:

        multi_source_type_claims.append(
            claim[
                "claim_id"
            ]
        )


# =========================================================
# Claim source / evidence multiplicity
# =========================================================

evidence_count_distribution = Counter(
    claim[
        "evidence_count"
    ]
    for claim in (
        profile_claims
    )
)


raw_claim_count_distribution = Counter(
    claim[
        "raw_claim_count"
    ]
    for claim in (
        profile_claims
    )
)


source_id_count_distribution = Counter(
    len(
        claim.get(
            "source_ids",
            []
        )
    )
    for claim in (
        profile_claims
    )
)


source_document_count_distribution = Counter(
    len(
        claim.get(
            "source_document_ids",
            []
        )
    )
    for claim in (
        profile_claims
    )
)


claims_with_multiple_evidence = sum(
    1
    for claim in (
        profile_claims
    )
    if (
        claim[
            "evidence_count"
        ]
        > 1
    )
)


claims_with_multiple_sources = sum(
    1
    for claim in (
        profile_claims
    )
    if (
        len(
            claim.get(
                "source_ids",
                []
            )
        )
        > 1
    )
)


claims_with_multiple_documents = sum(
    1
    for claim in (
        profile_claims
    )
    if (
        len(
            claim.get(
                "source_document_ids",
                []
            )
        )
        > 1
    )
)


# =========================================================
# Claim text length profile
# =========================================================

claim_text_lengths = [
    len(
        claim[
            "claim_text_fa"
        ]
    )
    for claim in (
        profile_claims
    )
]


claim_text_length_profile = (
    numeric_profile(
        claim_text_lengths
    )
)


claim_text_length_bands = Counter()


for length in (
    claim_text_lengths
):

    if length <= 40:

        band = (
            "LE_40"
        )

    elif length <= 80:

        band = (
            "41_TO_80"
        )

    elif length <= 160:

        band = (
            "81_TO_160"
        )

    elif length <= 320:

        band = (
            "161_TO_320"
        )

    elif length <= 640:

        band = (
            "321_TO_640"
        )

    else:

        band = (
            "GT_640"
        )


    claim_text_length_bands[
        band
    ] += 1


# =========================================================
# Evidence quality profile
# =========================================================

evidence_quote_lengths = []


evidence_missing_quote = 0
evidence_missing_locator = 0
evidence_missing_source_id = 0
evidence_missing_document_id = 0
evidence_missing_attribution = 0


for evidence in (
    profile_registry_evidence
):

    quote = evidence.get(
        "evidence_quote_exact"
    )


    if not is_nonempty_value(
        quote
    ):

        evidence_missing_quote += 1

    else:

        evidence_quote_lengths.append(
            len(
                quote
            )
        )


    if not is_nonempty_value(
        evidence.get(
            "locator"
        )
    ):

        evidence_missing_locator += 1


    if not is_nonempty_value(
        evidence.get(
            "source_id"
        )
    ):

        evidence_missing_source_id += 1


    if not is_nonempty_value(
        evidence.get(
            "document_id"
        )
    ):

        evidence_missing_document_id += 1


    if not is_nonempty_value(
        evidence.get(
            "attribution_snapshot"
        )
    ):

        evidence_missing_attribution += 1


assert (
    evidence_missing_quote
    == 0
), (
    "Validated Registry contains Evidence "
    "without exact quote."
)


assert (
    evidence_missing_source_id
    == 0
)


assert (
    evidence_missing_document_id
    == 0
)


evidence_quote_length_profile = (
    numeric_profile(
        evidence_quote_lengths
    )
)


# =========================================================
# Relation topology
# =========================================================

claim_relation_degree = Counter()

claim_relation_type_degree = {
    "SEMANTIC_DUPLICATE": Counter(),
    "SUBSUMES": Counter(),
    "PARTIAL_OVERLAP": Counter(),
}


graph_relations = [
    relation
    for relation in (
        profile_relations
    )
    if relation.get(
        "graph_edge_eligible"
    )
]


assert (
    len(
        graph_relations
    )
    == registry_counts[
        "graph_edge_eligible_claim_relations"
    ]
)


for relation in (
    graph_relations
):

    relation_type = relation[
        "relation_type"
    ]

    claim_id_a = relation[
        "claim_id_a"
    ]

    claim_id_b = relation[
        "claim_id_b"
    ]


    assert (
        claim_id_a
        in profile_claim_by_id
    )

    assert (
        claim_id_b
        in profile_claim_by_id
    )

    assert (
        claim_id_a
        != claim_id_b
    )


    claim_relation_degree[
        claim_id_a
    ] += 1

    claim_relation_degree[
        claim_id_b
    ] += 1


    assert (
        relation_type
        in claim_relation_type_degree
    )


    claim_relation_type_degree[
        relation_type
    ][
        claim_id_a
    ] += 1

    claim_relation_type_degree[
        relation_type
    ][
        claim_id_b
    ] += 1


claims_with_any_graph_relation = set(
    claim_relation_degree
)


claims_without_graph_relation = sorted(
    set(
        profile_claim_by_id
    )
    - claims_with_any_graph_relation
)


relation_degree_values = [
    claim_relation_degree.get(
        claim_id,
        0,
    )
    for claim_id in (
        profile_claim_by_id
    )
]


relation_degree_profile = numeric_profile(
    relation_degree_values
)


claims_with_semantic_duplicate = set(
    claim_relation_type_degree[
        "SEMANTIC_DUPLICATE"
    ]
)


claims_with_subsumes = set(
    claim_relation_type_degree[
        "SUBSUMES"
    ]
)


claims_with_partial_overlap = set(
    claim_relation_type_degree[
        "PARTIAL_OVERLAP"
    ]
)


# =========================================================
# Exact duplicate provenance profile
# =========================================================

exact_duplicate_collapsed_claims = [
    claim[
        "claim_id"
    ]
    for claim in (
        profile_claims
    )
    if claim.get(
        "exact_duplicate_collapsed"
    )
]


# =========================================================
# Structural failure counts
# =========================================================

structural_failure_reason_counts = Counter(
    reason
    for reasons in (
        structural_failures_by_claim.values()
    )
    for reason in reasons
)


# =========================================================
# Stage result
# =========================================================

claim_eligibility_profile_status = (
    "PASS"
)


golden_v2_eligibility_contract_readiness = (
    "READY_FOR_GOLDEN_V2_ELIGIBILITY_CONTRACT"
)


# =========================================================
# Result
# =========================================================

print(
    "=== NB12j REGISTRY CLAIM ELIGIBILITY PROFILE ==="
)

print()

print(
    "Canonical Claims:",
    len(
        profile_claims
    ),
)

print(
    "Structurally eligible for Golden review:",
    len(
        structurally_eligible_claim_ids
    ),
)

print(
    "Structurally ineligible:",
    len(
        structurally_ineligible_claim_ids
    ),
)

print(
    "Structural failure reasons:",
    sorted_counter_dict(
        structural_failure_reason_counts
    ),
)


print()

print(
    "=== CLAIM TYPES ==="
)

print(
    "Claim type membership counts:",
    sorted_counter_dict(
        claim_type_membership_counts
    ),
)

print(
    "Canonical singular claim_type counts:",
    sorted_counter_dict(
        canonical_singular_claim_type_counts
    ),
)

print(
    "Claims with multiple Claim Types:",
    len(
        multi_claim_type_claims
    ),
)


print()

print(
    "=== AUTHORITY / ROLE / SOURCE ==="
)

print(
    "Authority membership counts:",
    sorted_counter_dict(
        authority_membership_counts
    ),
)

print(
    "Discovery Role membership counts:",
    sorted_counter_dict(
        discovery_role_membership_counts
    ),
)

print(
    "Source Type membership counts:",
    sorted_counter_dict(
        source_type_membership_counts
    ),
)

print(
    "Multi-authority Claims:",
    len(
        multi_authority_claims
    ),
)

print(
    "Multi-role Claims:",
    len(
        multi_discovery_role_claims
    ),
)

print(
    "Multi-source-type Claims:",
    len(
        multi_source_type_claims
    ),
)


print()

print(
    "=== EVIDENCE / SOURCE MULTIPLICITY ==="
)

print(
    "Evidence count distribution:",
    sorted_counter_dict(
        evidence_count_distribution
    ),
)

print(
    "Raw Claim count distribution:",
    sorted_counter_dict(
        raw_claim_count_distribution
    ),
)

print(
    "Source ID count distribution:",
    sorted_counter_dict(
        source_id_count_distribution
    ),
)

print(
    "Source Document count distribution:",
    sorted_counter_dict(
        source_document_count_distribution
    ),
)

print(
    "Claims with >1 Evidence:",
    claims_with_multiple_evidence,
)

print(
    "Claims with >1 Source:",
    claims_with_multiple_sources,
)

print(
    "Claims with >1 Document:",
    claims_with_multiple_documents,
)


print()

print(
    "=== CLAIM TEXT LENGTH ==="
)

print(
    "Length profile:",
    claim_text_length_profile,
)

print(
    "Length bands:",
    sorted_counter_dict(
        claim_text_length_bands
    ),
)


print()

print(
    "=== VALIDATED EVIDENCE PROFILE ==="
)

print(
    "Registry Evidence occurrences:",
    len(
        profile_registry_evidence
    ),
)

print(
    "Evidence quote length profile:",
    evidence_quote_length_profile,
)

print(
    "Missing exact quote:",
    evidence_missing_quote,
)

print(
    "Missing locator:",
    evidence_missing_locator,
)

print(
    "Missing source_id:",
    evidence_missing_source_id,
)

print(
    "Missing document_id:",
    evidence_missing_document_id,
)

print(
    "Missing attribution snapshot:",
    evidence_missing_attribution,
)


print()

print(
    "=== RELATION TOPOLOGY ==="
)

print(
    "Graph-edge eligible Relations:",
    len(
        graph_relations
    ),
)

print(
    "Claims with any Graph Relation:",
    len(
        claims_with_any_graph_relation
    ),
)

print(
    "Claims without Graph Relation:",
    len(
        claims_without_graph_relation
    ),
)

print(
    "Relation degree profile:",
    relation_degree_profile,
)

print(
    "Claims with SEMANTIC_DUPLICATE relation:",
    len(
        claims_with_semantic_duplicate
    ),
)

print(
    "Claims with SUBSUMES relation:",
    len(
        claims_with_subsumes
    ),
)

print(
    "Claims with PARTIAL_OVERLAP relation:",
    len(
        claims_with_partial_overlap
    ),
)

print(
    "Canonical Claims with Exact-Duplicate provenance:",
    len(
        exact_duplicate_collapsed_claims
    ),
)


print()

print(
    "Final Golden Seed eligibility decided:",
    False,
)

print(
    "Fixed Golden Seed target invented:",
    False,
)

print(
    "Files modified by this Cell:",
    0,
)

print(
    "API calls performed:",
    0,
)


print()

print(
    "Next stage:",
    "GOLDEN_V2_ELIGIBILITY_CONTRACT",
)

print(
    "Readiness:",
    golden_v2_eligibility_contract_readiness,
)


print()

print(
    "CODE CELL 06 GATE:",
    claim_eligibility_profile_status,
)
````

<div style="direction: rtl; white-space: normal; line-height: 1.7;">

<h2>Freeze کردن Golden v2 Claim Eligibility Policy</h2>

<p>
پروفایل Registry نشان داد که هر 4006 Canonical Claim از نظر ساختاری
برای ورود به بررسی Golden مناسب است و هیچ Claim به علت فقدان Evidence،
Source، Document، Authority، Claim Type یا Validation Status رد نشده است.
</p>

<p>
با این حال، Structural Eligibility با Final Golden Seed Eligibility یکسان نیست.
یک Claim معتبر ممکن است از نظر معنایی با Claim دیگری Duplicate، Subsumed
یا دارای Partial Overlap باشد و بنابراین نباید بدون کنترل Redundancy مستقیماً
به یک Seed مستقل تبدیل شود.
</p>

<h3>اصل اساسی</h3>

<p>
هیچ Claim معتبر صرفاً به دلیل نوع Claim، Source Type، طول متن،
Primary/Secondary بودن یا داشتن Relation حذف نمی‌شود.
این ویژگی‌ها برای Routing، Review و Coverage استفاده می‌شوند.
</p>

<h3>Authority Policy</h3>

<ul>
<li><code>CORE_PRIMARY_KNOWLEDGE</code> وارد مسیر Standard Golden Review می‌شود.</li>
<li><code>SUPERVISED_SECONDARY_KNOWLEDGE</code> می‌تواند وارد Golden شود، اما Attribution آن باید حفظ شود.</li>
<li>Secondary Knowledge حق Override کردن Primary Evidence را ندارد.</li>
</ul>

<h3>Relation Policy</h3>

<ul>
<li><code>SEMANTIC_DUPLICATE</code> باعث حذف خودکار Claim نمی‌شود، اما Claim وارد Redundancy Review می‌شود.</li>
<li><code>SUBSUMES</code> باعث Merge خودکار نمی‌شود؛ Claimهای Parent و Child باید از نظر استقلال سؤال بررسی شوند.</li>
<li><code>PARTIAL_OVERLAP</code> باعث حذف نمی‌شود، اما برای جلوگیری از Question Redundancy علامت‌گذاری می‌شود.</li>
<li>Exact Duplicateها قبلاً در Registry Canonicalize شده‌اند و دیگر Seed مستقل جداگانه ایجاد نمی‌کنند.</li>
</ul>

<h3>Claim Type Policy</h3>

<p>
تمام Claim Typeهای مشاهده‌شده در Registry مجاز به ورود به Golden Suitability Review هستند.
در این مرحله هیچ Claim Type با Threshold دلخواه حذف نمی‌شود.
</p>

<h3>Text Length Policy</h3>

<p>
طول Claim Text یک Diagnostic Feature است و Hard Exclusion ایجاد نمی‌کند.
Claim کوتاه یا بلند فقط در مرحله Question Generation و Quality Review
باید از نظر قابلیت ساخت سؤال مستقل بررسی شود.
</p>

<h3>Final Seed Requirement</h3>

<p>
یک Claim فقط زمانی Final Golden Seed محسوب خواهد شد که علاوه بر Structural Eligibility،
بتوان از آن یک سؤال مستقل، غیرمبهم، Evidence-Grounded و قابل پاسخ از Evidence واقعی ساخت.
این تصمیم در مراحل بعد انجام می‌شود و در این مرحله هیچ تعداد Seed هدف اختراع نمی‌شود.
</p>

<h3>نتیجه این مرحله</h3>

<p>
یک Policy deterministic با Hash مشخص ساخته می‌شود و مرحله بعد بر اساس آن
برای هر 4006 Claim یک Eligibility Manifest مستقل تولید خواهد کرد.
</p>

</div>

````python
# Code Cell 07

assert (
    claim_eligibility_profile_status
    == "PASS"
), (
    "Code Cell 06 must PASS before "
    "freezing Golden v2 Eligibility Policy."
)

assert (
    golden_v2_eligibility_contract_readiness
    == "READY_FOR_GOLDEN_V2_ELIGIBILITY_CONTRACT"
)


# =========================================================
# Configuration
# =========================================================

GOLDEN_V2_ELIGIBILITY_POLICY_VERSION = (
    "golden_v2_claim_eligibility_policy_v1"
)


# =========================================================
# Actual observed Registry classes
# =========================================================

observed_claim_types = sorted(
    claim_type_membership_counts
)

observed_authority_classes = sorted(
    authority_membership_counts
)

observed_discovery_roles = sorted(
    discovery_role_membership_counts
)

observed_source_types = sorted(
    source_type_membership_counts
)


assert (
    set(
        observed_claim_types
    )
    == {
        "CAUSAL",
        "CLASSIFICATION",
        "COMPARISON",
        "CONDITION_OR_CONSTRAINT",
        "DEFINITION",
        "NUMERIC_OR_QUANTITATIVE",
        "OTHER_ATOMIC",
        "PROCEDURE_OR_ACTION",
        "RELATION",
        "RULE_OR_PRINCIPLE",
    }
)


assert (
    set(
        observed_authority_classes
    )
    == {
        "CORE_PRIMARY_KNOWLEDGE",
        "SUPERVISED_SECONDARY_KNOWLEDGE",
    }
)


assert (
    set(
        observed_discovery_roles
    )
    == {
        "PRIMARY_CLAIM_DISCOVERY",
        "SECONDARY_CLAIM_DISCOVERY_WITH_ATTRIBUTION",
    }
)


assert (
    set(
        observed_source_types
    )
    == {
        "book",
        "transcript",
        "website",
    }
)


# =========================================================
# Relation-review populations
# =========================================================

relation_review_claim_ids = (
    claims_with_semantic_duplicate
    | claims_with_subsumes
    | claims_with_partial_overlap
)


relation_isolated_claim_ids = (
    set(
        profile_claim_by_id
    )
    - relation_review_claim_ids
)


assert (
    len(
        relation_review_claim_ids
    )
    == len(
        claims_with_any_graph_relation
    )
)


assert (
    len(
        relation_isolated_claim_ids
    )
    == len(
        claims_without_graph_relation
    )
)


# =========================================================
# Authority lanes
# =========================================================

primary_claim_ids = {
    claim[
        "claim_id"
    ]
    for claim in (
        profile_claims
    )
    if (
        claim[
            "authority_classes"
        ]
        == [
            "CORE_PRIMARY_KNOWLEDGE"
        ]
    )
}


secondary_claim_ids = {
    claim[
        "claim_id"
    ]
    for claim in (
        profile_claims
    )
    if (
        claim[
            "authority_classes"
        ]
        == [
            "SUPERVISED_SECONDARY_KNOWLEDGE"
        ]
    )
}


assert (
    len(
        primary_claim_ids
    )
    == 2797
)


assert (
    len(
        secondary_claim_ids
    )
    == 1209
)


assert (
    primary_claim_ids
    | secondary_claim_ids
    == set(
        profile_claim_by_id
    )
)


assert not (
    primary_claim_ids
    & secondary_claim_ids
)


# =========================================================
# Review flags
# =========================================================

eligibility_review_flag_definitions = {
    "ATTRIBUTION_REQUIRED": (
        "Claim is supervised secondary knowledge and "
        "must preserve attribution in Golden generation."
    ),

    "SEMANTIC_DUPLICATE_REVIEW": (
        "Claim participates in at least one adjudicated "
        "SEMANTIC_DUPLICATE relation."
    ),

    "SUBSUMPTION_REVIEW": (
        "Claim participates in at least one adjudicated "
        "SUBSUMES relation."
    ),

    "PARTIAL_OVERLAP_REVIEW": (
        "Claim participates in at least one adjudicated "
        "PARTIAL_OVERLAP relation."
    ),

    "MULTI_CLAIM_TYPE_REVIEW": (
        "Canonical Claim preserves more than one "
        "Claim Type."
    ),

    "MULTI_EVIDENCE": (
        "Claim has more than one validated Evidence "
        "occurrence."
    ),

    "MULTI_SOURCE": (
        "Claim is supported by more than one source_id."
    ),

    "MULTI_DOCUMENT": (
        "Claim is supported by more than one document."
    ),

    "EXACT_DUPLICATE_PROVENANCE": (
        "Canonical Claim collapsed more than one Raw Claim "
        "through exact-text canonicalization."
    ),

    "SHORT_TEXT_DIAGNOSTIC": (
        "Claim text has 40 or fewer characters. "
        "This is diagnostic only and not a rejection."
    ),
}


# =========================================================
# Frozen eligibility policy
# =========================================================

golden_v2_eligibility_policy = {
    "policy_version": (
        GOLDEN_V2_ELIGIBILITY_POLICY_VERSION
    ),

    "registry_population": (
        len(
            profile_claims
        )
    ),

    "structurally_eligible_population": (
        len(
            structurally_eligible_claim_ids
        )
    ),

    "structurally_ineligible_population": (
        len(
            structurally_ineligible_claim_ids
        )
    ),

    "structural_requirement": {
        "validated_claim_required": True,
        "claim_text_required": True,
        "evidence_required": True,
        "source_id_required": True,
        "document_id_required": True,
        "source_type_required": True,
        "authority_class_required": True,
        "claim_discovery_role_required": True,
        "claim_type_required": True,
    },

    "authority_policy": {
        "CORE_PRIMARY_KNOWLEDGE": (
            "STANDARD_GOLDEN_SUITABILITY_REVIEW"
        ),

        "SUPERVISED_SECONDARY_KNOWLEDGE": (
            "ATTRIBUTED_GOLDEN_SUITABILITY_REVIEW"
        ),

        "secondary_may_override_primary": False,

        "secondary_requires_attribution": True,
    },

    "claim_type_policy": {
        "observed_allowed_claim_types": (
            observed_claim_types
        ),

        "hard_excluded_claim_types": [],

        "claim_type_alone_can_reject_seed": False,
    },

    "source_policy": {
        "observed_source_types": (
            observed_source_types
        ),

        "hard_excluded_source_types": [],

        "source_type_alone_can_reject_seed": False,
    },

    "text_length_policy": {
        "hard_minimum_characters": None,

        "hard_maximum_characters": None,

        "length_alone_can_reject_seed": False,

        "short_text_diagnostic_threshold_chars": 40,
    },

    "relation_policy": {
        "EXACT_DUPLICATE": (
            "ALREADY_CANONICALIZED_NOT_A_SEPARATE_SEED"
        ),

        "SEMANTIC_DUPLICATE": (
            "PRESERVE_CLAIMS_REQUIRE_REDUNDANCY_REVIEW"
        ),

        "SUBSUMES": (
            "PRESERVE_CLAIMS_REQUIRE_INDEPENDENCE_REVIEW"
        ),

        "PARTIAL_OVERLAP": (
            "PRESERVE_CLAIMS_REQUIRE_QUESTION_REDUNDANCY_REVIEW"
        ),

        "automatic_semantic_duplicate_rejection": False,

        "automatic_subsumption_rejection": False,

        "automatic_partial_overlap_rejection": False,
    },

    "review_flags": (
        eligibility_review_flag_definitions
    ),

    "final_seed_requirements": {
        "structurally_eligible": True,

        "canonical_claim_id_required": True,

        "validated_evidence_required": True,

        "traceable_source_required": True,

        "question_must_be_independent": True,

        "question_must_be_unambiguous": True,

        "question_must_be_answerable_from_evidence": True,

        "answer_must_be_grounded_in_evidence": True,

        "must_not_require_unprovided_external_context": True,

        "semantic_redundancy_must_be_controlled": True,

        "question_redundancy_must_be_controlled": True,
    },

    "seed_count_policy": {
        "fixed_target": None,

        "fixed_target_invented": False,

        "derive_from_actual_eligibility": True,

        "derive_from_redundancy_control": True,

        "derive_from_coverage_requirements": True,
    },

    "selection_state": (
        "STRUCTURAL_REVIEW_READY_FINAL_SEED_SELECTION_NOT_STARTED"
    ),
}


# =========================================================
# Deterministic policy hash
#
# Policy is carried into the Claim Eligibility Manifest.
# No additional unplanned artifact is written here.
# =========================================================

golden_v2_eligibility_policy_text = (
    json.dumps(
        golden_v2_eligibility_policy,
        ensure_ascii=False,
        sort_keys=True,
        separators=(
            ",",
            ":",
        ),
    )
)


golden_v2_eligibility_policy_sha256 = (
    hashlib.sha256(
        golden_v2_eligibility_policy_text.encode(
            "utf-8"
        )
    ).hexdigest()
)


# =========================================================
# Review-flag population diagnostics
# =========================================================

eligibility_flag_population_counts = {
    "ATTRIBUTION_REQUIRED": (
        len(
            secondary_claim_ids
        )
    ),

    "SEMANTIC_DUPLICATE_REVIEW": (
        len(
            claims_with_semantic_duplicate
        )
    ),

    "SUBSUMPTION_REVIEW": (
        len(
            claims_with_subsumes
        )
    ),

    "PARTIAL_OVERLAP_REVIEW": (
        len(
            claims_with_partial_overlap
        )
    ),

    "MULTI_CLAIM_TYPE_REVIEW": (
        len(
            multi_claim_type_claims
        )
    ),

    "MULTI_EVIDENCE": (
        claims_with_multiple_evidence
    ),

    "MULTI_SOURCE": (
        claims_with_multiple_sources
    ),

    "MULTI_DOCUMENT": (
        claims_with_multiple_documents
    ),

    "EXACT_DUPLICATE_PROVENANCE": (
        len(
            exact_duplicate_collapsed_claims
        )
    ),

    "SHORT_TEXT_DIAGNOSTIC": (
        claim_text_length_bands[
            "LE_40"
        ]
    ),
}


# =========================================================
# Contract validation
# =========================================================

assert (
    golden_v2_eligibility_policy[
        "registry_population"
    ]
    == 4006
)


assert (
    golden_v2_eligibility_policy[
        "structurally_eligible_population"
    ]
    == 4006
)


assert (
    golden_v2_eligibility_policy[
        "structurally_ineligible_population"
    ]
    == 0
)


assert (
    golden_v2_eligibility_policy[
        "claim_type_policy"
    ][
        "hard_excluded_claim_types"
    ]
    == []
)


assert (
    golden_v2_eligibility_policy[
        "source_policy"
    ][
        "hard_excluded_source_types"
    ]
    == []
)


assert (
    golden_v2_eligibility_policy[
        "relation_policy"
    ][
        "automatic_semantic_duplicate_rejection"
    ]
    is False
)


assert (
    golden_v2_eligibility_policy[
        "seed_count_policy"
    ][
        "fixed_target"
    ]
    is None
)


# =========================================================
# Gate
# =========================================================

golden_v2_eligibility_policy_status = (
    "PASS"
)


claim_eligibility_manifest_readiness = (
    "READY_FOR_CLAIM_ELIGIBILITY_MANIFEST"
)


# =========================================================
# Result
# =========================================================

print(
    "=== NB12j GOLDEN V2 ELIGIBILITY POLICY ==="
)

print()

print(
    "Registry Claims:",
    len(
        profile_claims
    ),
)

print(
    "Structurally eligible:",
    len(
        structurally_eligible_claim_ids
    ),
)

print(
    "Hard structural rejects:",
    len(
        structurally_ineligible_claim_ids
    ),
)


print()

print(
    "Primary review lane:",
    len(
        primary_claim_ids
    ),
)

print(
    "Secondary attributed review lane:",
    len(
        secondary_claim_ids
    ),
)


print()

print(
    "Claims involved in any semantic relation review:",
    len(
        relation_review_claim_ids
    ),
)

print(
    "Claims with no adjudicated semantic relation:",
    len(
        relation_isolated_claim_ids
    ),
)


print()

print(
    "Review flag populations:"
)

for (
    flag_name,
    flag_count,
) in sorted(
    eligibility_flag_population_counts.items()
):

    print(
        "  ",
        flag_name,
        ":",
        flag_count,
    )


print()

print(
    "Hard-excluded Claim Types:",
    golden_v2_eligibility_policy[
        "claim_type_policy"
    ][
        "hard_excluded_claim_types"
    ],
)

print(
    "Hard-excluded Source Types:",
    golden_v2_eligibility_policy[
        "source_policy"
    ][
        "hard_excluded_source_types"
    ],
)

print(
    "Automatic Semantic-Duplicate rejection:",
    False,
)

print(
    "Automatic SUBSUMES rejection:",
    False,
)

print(
    "Automatic Partial-Overlap rejection:",
    False,
)


print()

print(
    "Fixed Golden Seed target:",
    golden_v2_eligibility_policy[
        "seed_count_policy"
    ][
        "fixed_target"
    ],
)

print(
    "Final Seed Selection started:",
    False,
)


print()

print(
    "Eligibility Policy version:",
    GOLDEN_V2_ELIGIBILITY_POLICY_VERSION,
)

print(
    "Eligibility Policy SHA-256:",
    golden_v2_eligibility_policy_sha256,
)

print(
    "Files modified by this Cell:",
    0,
)

print(
    "API calls performed:",
    0,
)


print()

print(
    "Next stage:",
    "CLAIM_ELIGIBILITY_MANIFEST",
)

print(
    "Readiness:",
    claim_eligibility_manifest_readiness,
)


print()

print(
    "CODE CELL 07 GATE:",
    golden_v2_eligibility_policy_status,
)
````

````python
# Code Cell 08

assert (
    golden_v2_eligibility_policy_status
    == "PASS"
), (
    "Code Cell 07 must PASS before "
    "Claim Eligibility Manifest generation."
)

assert (
    claim_eligibility_manifest_readiness
    == "READY_FOR_CLAIM_ELIGIBILITY_MANIFEST"
)


# =========================================================
# Configuration
# =========================================================

CLAIM_ELIGIBILITY_MANIFEST_VERSION = (
    "golden_v2_claim_eligibility_manifest_v1"
)


claim_eligibility_manifest_path = (
    golden_v2_artifact_root
    / golden_v2_artifact_family[
        "claim_eligibility_manifest"
    ]
)


# =========================================================
# Relation lookup per Claim
# =========================================================

relation_counts_by_claim = {
    claim_id: {
        "SEMANTIC_DUPLICATE": 0,
        "SUBSUMES": 0,
        "PARTIAL_OVERLAP": 0,
    }
    for claim_id in (
        profile_claim_by_id
    )
}


relation_ids_by_claim = {
    claim_id: {
        "SEMANTIC_DUPLICATE": [],
        "SUBSUMES": [],
        "PARTIAL_OVERLAP": [],
    }
    for claim_id in (
        profile_claim_by_id
    )
}


for relation in (
    graph_relations
):

    relation_type = relation[
        "relation_type"
    ]

    relation_id = relation[
        "relation_id"
    ]

    claim_id_a = relation[
        "claim_id_a"
    ]

    claim_id_b = relation[
        "claim_id_b"
    ]


    for claim_id in [
        claim_id_a,
        claim_id_b,
    ]:

        relation_counts_by_claim[
            claim_id
        ][
            relation_type
        ] += 1


        relation_ids_by_claim[
            claim_id
        ][
            relation_type
        ].append(
            relation_id
        )


# =========================================================
# Evidence lookup
# =========================================================

manifest_evidence_by_id = {
    evidence[
        "evidence_id"
    ]: evidence
    for evidence in (
        profile_registry_evidence
    )
}


assert (
    len(
        manifest_evidence_by_id
    )
    == 4125
)


# =========================================================
# Build one deterministic Manifest record per Claim
# =========================================================

claim_eligibility_manifest_records = []


for claim in sorted(
    profile_claims,
    key=lambda item: (
        item[
            "claim_id"
        ]
    ),
):

    claim_id = claim[
        "claim_id"
    ]


    # -----------------------------------------------------
    # Structural eligibility
    # -----------------------------------------------------

    structural_failures = (
        structural_failures_by_claim.get(
            claim_id,
            [],
        )
    )


    structurally_eligible = (
        len(
            structural_failures
        )
        == 0
    )


    assert (
        structurally_eligible
        is True
    )


    # -----------------------------------------------------
    # Authority lane
    # -----------------------------------------------------

    authority_classes = (
        claim[
            "authority_classes"
        ]
    )


    assert (
        len(
            authority_classes
        )
        == 1
    )


    authority_class = (
        authority_classes[
            0
        ]
    )


    if (
        authority_class
        == "CORE_PRIMARY_KNOWLEDGE"
    ):

        authority_lane = (
            "PRIMARY_STANDARD_REVIEW"
        )


    elif (
        authority_class
        == "SUPERVISED_SECONDARY_KNOWLEDGE"
    ):

        authority_lane = (
            "SECONDARY_ATTRIBUTED_REVIEW"
        )


    else:

        raise AssertionError(
            "Unexpected authority class: "
            f"{authority_class}"
        )


    # -----------------------------------------------------
    # Relation profile
    # -----------------------------------------------------

    relation_counts = (
        relation_counts_by_claim[
            claim_id
        ]
    )


    total_relation_degree = sum(
        relation_counts.values()
    )


    # -----------------------------------------------------
    # Review flags
    # -----------------------------------------------------

    review_flags = []


    if (
        authority_class
        == "SUPERVISED_SECONDARY_KNOWLEDGE"
    ):

        review_flags.append(
            "ATTRIBUTION_REQUIRED"
        )


    if (
        relation_counts[
            "SEMANTIC_DUPLICATE"
        ]
        > 0
    ):

        review_flags.append(
            "SEMANTIC_DUPLICATE_REVIEW"
        )


    if (
        relation_counts[
            "SUBSUMES"
        ]
        > 0
    ):

        review_flags.append(
            "SUBSUMPTION_REVIEW"
        )


    if (
        relation_counts[
            "PARTIAL_OVERLAP"
        ]
        > 0
    ):

        review_flags.append(
            "PARTIAL_OVERLAP_REVIEW"
        )


    if (
        len(
            claim[
                "claim_types"
            ]
        )
        > 1
    ):

        review_flags.append(
            "MULTI_CLAIM_TYPE_REVIEW"
        )


    if (
        claim[
            "evidence_count"
        ]
        > 1
    ):

        review_flags.append(
            "MULTI_EVIDENCE"
        )


    if (
        len(
            claim[
                "source_ids"
            ]
        )
        > 1
    ):

        review_flags.append(
            "MULTI_SOURCE"
        )


    if (
        len(
            claim[
                "source_document_ids"
            ]
        )
        > 1
    ):

        review_flags.append(
            "MULTI_DOCUMENT"
        )


    if (
        claim[
            "exact_duplicate_collapsed"
        ]
        is True
    ):

        review_flags.append(
            "EXACT_DUPLICATE_PROVENANCE"
        )


    claim_text_length = len(
        claim[
            "claim_text_fa"
        ]
    )


    if (
        claim_text_length
        <= 40
    ):

        review_flags.append(
            "SHORT_TEXT_DIAGNOSTIC"
        )


    review_flags = sorted(
        review_flags
    )


    # -----------------------------------------------------
    # Evidence profile for this Claim
    # -----------------------------------------------------

    claim_evidence = [
        manifest_evidence_by_id[
            evidence_id
        ]
        for evidence_id in (
            claim[
                "evidence_ids"
            ]
        )
    ]


    assert (
        len(
            claim_evidence
        )
        == claim[
            "evidence_count"
        ]
    )


    evidence_quote_lengths_for_claim = [
        len(
            evidence[
                "evidence_quote_exact"
            ]
        )
        for evidence in (
            claim_evidence
        )
    ]


    assert all(
        is_nonempty_value(
            evidence[
                "evidence_quote_exact"
            ]
        )
        for evidence in (
            claim_evidence
        )
    )


    assert all(
        is_nonempty_value(
            evidence[
                "locator"
            ]
        )
        for evidence in (
            claim_evidence
        )
    )


    assert all(
        is_nonempty_value(
            evidence[
                "source_id"
            ]
        )
        for evidence in (
            claim_evidence
        )
    )


    assert all(
        is_nonempty_value(
            evidence[
                "document_id"
            ]
        )
        for evidence in (
            claim_evidence
        )
    )


    # -----------------------------------------------------
    # Manifest record
    # -----------------------------------------------------

    claim_eligibility_manifest_records.append({
        "manifest_version": (
            CLAIM_ELIGIBILITY_MANIFEST_VERSION
        ),

        "eligibility_policy_version": (
            GOLDEN_V2_ELIGIBILITY_POLICY_VERSION
        ),

        "eligibility_policy_sha256": (
            golden_v2_eligibility_policy_sha256
        ),

        "claim_id": (
            claim_id
        ),

        "claim_text_fa": (
            claim[
                "claim_text_fa"
            ]
        ),

        "claim_text_sha256": (
            claim[
                "claim_text_sha256"
            ]
        ),

        "claim_text_length_chars": (
            claim_text_length
        ),

        "claim_type": (
            claim[
                "claim_type"
            ]
        ),

        "claim_types": (
            claim[
                "claim_types"
            ]
        ),

        "authority_class": (
            authority_class
        ),

        "authority_lane": (
            authority_lane
        ),

        "claim_discovery_roles": (
            claim[
                "claim_discovery_roles"
            ]
        ),

        "source_types": (
            claim[
                "source_types"
            ]
        ),

        "source_ids": (
            claim[
                "source_ids"
            ]
        ),

        "source_document_ids": (
            claim[
                "source_document_ids"
            ]
        ),

        "evidence_ids": (
            claim[
                "evidence_ids"
            ]
        ),

        "evidence_count": (
            claim[
                "evidence_count"
            ]
        ),

        "evidence_quote_length_min": (
            min(
                evidence_quote_lengths_for_claim
            )
        ),

        "evidence_quote_length_max": (
            max(
                evidence_quote_lengths_for_claim
            )
        ),

        "raw_claim_ids": (
            claim[
                "raw_claim_ids"
            ]
        ),

        "raw_claim_count": (
            claim[
                "raw_claim_count"
            ]
        ),

        "exact_duplicate_collapsed": (
            claim[
                "exact_duplicate_collapsed"
            ]
        ),

        "relation_profile": {
            "total_graph_relation_degree": (
                total_relation_degree
            ),

            "semantic_duplicate_count": (
                relation_counts[
                    "SEMANTIC_DUPLICATE"
                ]
            ),

            "subsumes_count": (
                relation_counts[
                    "SUBSUMES"
                ]
            ),

            "partial_overlap_count": (
                relation_counts[
                    "PARTIAL_OVERLAP"
                ]
            ),

            "semantic_duplicate_relation_ids": sorted(
                relation_ids_by_claim[
                    claim_id
                ][
                    "SEMANTIC_DUPLICATE"
                ]
            ),

            "subsumes_relation_ids": sorted(
                relation_ids_by_claim[
                    claim_id
                ][
                    "SUBSUMES"
                ]
            ),

            "partial_overlap_relation_ids": sorted(
                relation_ids_by_claim[
                    claim_id
                ][
                    "PARTIAL_OVERLAP"
                ]
            ),
        },

        "review_flags": (
            review_flags
        ),

        "review_flag_count": (
            len(
                review_flags
            )
        ),

        "structural_eligibility": (
            "STRUCTURALLY_ELIGIBLE_FOR_GOLDEN_REVIEW"
        ),

        "structural_failures": (
            structural_failures
        ),

        "final_seed_decision": (
            "NOT_YET_DECIDED"
        ),

        "final_seed_selected": (
            False
        ),

        "question_generation_started": (
            False
        ),
    })


# =========================================================
# Structural validation
# =========================================================

assert (
    len(
        claim_eligibility_manifest_records
    )
    == 4006
)


manifest_claim_ids = [
    record[
        "claim_id"
    ]
    for record in (
        claim_eligibility_manifest_records
    )
]


assert (
    len(
        manifest_claim_ids
    )
    == len(
        set(
            manifest_claim_ids
        )
    )
)


assert (
    set(
        manifest_claim_ids
    )
    == set(
        profile_claim_by_id
    )
)


assert all(
    record[
        "structural_eligibility"
    ]
    == "STRUCTURALLY_ELIGIBLE_FOR_GOLDEN_REVIEW"
    for record in (
        claim_eligibility_manifest_records
    )
)


assert all(
    record[
        "final_seed_decision"
    ]
    == "NOT_YET_DECIDED"
    for record in (
        claim_eligibility_manifest_records
    )
)


assert all(
    record[
        "question_generation_started"
    ]
    is False
    for record in (
        claim_eligibility_manifest_records
    )
)


# =========================================================
# Recalculate flag populations from Manifest
# =========================================================

manifest_flag_counts = Counter(
    flag
    for record in (
        claim_eligibility_manifest_records
    )
    for flag in (
        record[
            "review_flags"
        ]
    )
)


assert (
    dict(
        sorted(
            manifest_flag_counts.items()
        )
    )
    == dict(
        sorted(
            eligibility_flag_population_counts.items()
        )
    )
)


# =========================================================
# Lane validation
# =========================================================

manifest_authority_lane_counts = Counter(
    record[
        "authority_lane"
    ]
    for record in (
        claim_eligibility_manifest_records
    )
)


assert (
    manifest_authority_lane_counts[
        "PRIMARY_STANDARD_REVIEW"
    ]
    == 2797
)


assert (
    manifest_authority_lane_counts[
        "SECONDARY_ATTRIBUTED_REVIEW"
    ]
    == 1209
)


# =========================================================
# Relation coverage validation
# =========================================================

manifest_relation_review_claims = {
    record[
        "claim_id"
    ]
    for record in (
        claim_eligibility_manifest_records
    )
    if (
        record[
            "relation_profile"
        ][
            "total_graph_relation_degree"
        ]
        > 0
    )
}


assert (
    manifest_relation_review_claims
    == relation_review_claim_ids
)


assert (
    len(
        manifest_relation_review_claims
    )
    == 1263
)


# =========================================================
# Evidence preservation validation
# =========================================================

manifest_evidence_ids = {
    evidence_id
    for record in (
        claim_eligibility_manifest_records
    )
    for evidence_id in (
        record[
            "evidence_ids"
        ]
    )
}


assert (
    manifest_evidence_ids
    == registry_evidence_ids
)


manifest_evidence_occurrences = sum(
    record[
        "evidence_count"
    ]
    for record in (
        claim_eligibility_manifest_records
    )
)


assert (
    manifest_evidence_occurrences
    == 4125
)


# =========================================================
# Serialize frozen Manifest
# =========================================================

claim_eligibility_manifest_text = "".join(
    json.dumps(
        record,
        ensure_ascii=False,
        sort_keys=True,
    )
    + "\n"
    for record in (
        claim_eligibility_manifest_records
    )
)


claim_eligibility_manifest_sha256 = (
    hashlib.sha256(
        claim_eligibility_manifest_text.encode(
            "utf-8"
        )
    ).hexdigest()
)


claim_eligibility_manifest_write_status = (
    publish_frozen_text(
        claim_eligibility_manifest_path,
        claim_eligibility_manifest_text,
    )
)


# =========================================================
# Independent disk revalidation
# =========================================================

disk_claim_eligibility_manifest = (
    load_jsonl_records(
        claim_eligibility_manifest_path
    )
)


assert (
    disk_claim_eligibility_manifest
    == claim_eligibility_manifest_records
)


assert (
    sha256_file(
        claim_eligibility_manifest_path
    )
    == claim_eligibility_manifest_sha256
)


# =========================================================
# Gate
# =========================================================

claim_eligibility_manifest_status = (
    "PASS"
)


seed_suitability_discovery_readiness = (
    "READY_FOR_SEED_SUITABILITY_DISCOVERY"
)


# =========================================================
# Result
# =========================================================

print(
    "=== NB12j CLAIM ELIGIBILITY MANIFEST ==="
)

print()

print(
    "Manifest records:",
    len(
        claim_eligibility_manifest_records
    ),
)

print(
    "Unique Claim IDs:",
    len(
        set(
            manifest_claim_ids
        )
    ),
)

print(
    "Structurally eligible:",
    sum(
        1
        for record in (
            claim_eligibility_manifest_records
        )
        if (
            record[
                "structural_eligibility"
            ]
            == "STRUCTURALLY_ELIGIBLE_FOR_GOLDEN_REVIEW"
        )
    ),
)


print()

print(
    "Authority lanes:",
    dict(
        sorted(
            manifest_authority_lane_counts.items()
        )
    ),
)


print()

print(
    "Claims requiring semantic relation review:",
    len(
        manifest_relation_review_claims
    ),
)

print(
    "Claims without semantic relation review:",
    (
        len(
            claim_eligibility_manifest_records
        )
        - len(
            manifest_relation_review_claims
        )
    ),
)


print()

print(
    "Review flag counts:",
    dict(
        sorted(
            manifest_flag_counts.items()
        )
    ),
)


print()

print(
    "Evidence occurrences preserved:",
    manifest_evidence_occurrences,
)

print(
    "Unique Evidence IDs preserved:",
    len(
        manifest_evidence_ids
    ),
)


print()

print(
    "Final Seed decisions made:",
    sum(
        1
        for record in (
            claim_eligibility_manifest_records
        )
        if (
            record[
                "final_seed_decision"
            ]
            != "NOT_YET_DECIDED"
        )
    ),
)

print(
    "Question generation started:",
    any(
        record[
            "question_generation_started"
        ]
        for record in (
            claim_eligibility_manifest_records
        )
    ),
)


print()

print(
    "Eligibility Manifest:",
    claim_eligibility_manifest_path.relative_to(
        project_root
    ),
)

print(
    "Eligibility Manifest SHA-256:",
    claim_eligibility_manifest_sha256,
)

print(
    "Write status:",
    claim_eligibility_manifest_write_status,
)


print()

print(
    "API calls performed:",
    0,
)

print(
    "Next stage:",
    "SEED_SUITABILITY_DISCOVERY",
)

print(
    "Readiness:",
    seed_suitability_discovery_readiness,
)


print()

print(
    "CODE CELL 08 GATE:",
    claim_eligibility_manifest_status,
)
````

````python
# Code Cell 09

assert (
    claim_eligibility_manifest_status
    == "PASS"
), (
    "Code Cell 08 must PASS before "
    "Seed Suitability Discovery."
)

assert (
    seed_suitability_discovery_readiness
    == "READY_FOR_SEED_SUITABILITY_DISCOVERY"
)


# =========================================================
# Graph helper
# =========================================================

def connected_components_from_edges(
    node_ids,
    edges,
):

    adjacency = defaultdict(
        set
    )


    for (
        node_a,
        node_b,
    ) in edges:

        assert (
            node_a
            in node_ids
        )

        assert (
            node_b
            in node_ids
        )

        assert (
            node_a
            != node_b
        )


        adjacency[
            node_a
        ].add(
            node_b
        )

        adjacency[
            node_b
        ].add(
            node_a
        )


    components = []

    visited = set()


    for start_node in sorted(
        adjacency
    ):

        if (
            start_node
            in visited
        ):
            continue


        stack = [
            start_node
        ]

        component = []


        while stack:

            current = stack.pop()


            if (
                current
                in visited
            ):
                continue


            visited.add(
                current
            )

            component.append(
                current
            )


            for neighbor in sorted(
                adjacency[
                    current
                ],
                reverse=True,
            ):

                if (
                    neighbor
                    not in visited
                ):

                    stack.append(
                        neighbor
                    )


        components.append(
            sorted(
                component
            )
        )


    components = sorted(
        components,
        key=lambda component: (
            -len(
                component
            ),
            component[
                0
            ],
        ),
    )


    return components


# =========================================================
# Split adjudicated Graph Relations by type
# =========================================================

semantic_duplicate_edges = []

subsumes_edges = []

partial_overlap_edges = []


for relation in (
    graph_relations
):

    relation_type = relation[
        "relation_type"
    ]

    claim_id_a = relation[
        "claim_id_a"
    ]

    claim_id_b = relation[
        "claim_id_b"
    ]


    if (
        relation_type
        == "SEMANTIC_DUPLICATE"
    ):

        semantic_duplicate_edges.append(
            (
                claim_id_a,
                claim_id_b,
            )
        )


    elif (
        relation_type
        == "SUBSUMES"
    ):

        subsumes_edges.append(
            (
                claim_id_a,
                claim_id_b,
            )
        )


    elif (
        relation_type
        == "PARTIAL_OVERLAP"
    ):

        partial_overlap_edges.append(
            (
                claim_id_a,
                claim_id_b,
            )
        )


    else:

        raise AssertionError(
            "Unexpected graph relation type: "
            f"{relation_type}"
        )


assert (
    len(
        semantic_duplicate_edges
    )
    == 233
)


assert (
    len(
        subsumes_edges
    )
    == 544
)


assert (
    len(
        partial_overlap_edges
    )
    == 1638
)


# =========================================================
# Semantic Duplicate connected components
# =========================================================

all_claim_ids = set(
    profile_claim_by_id
)


semantic_duplicate_components = (
    connected_components_from_edges(
        all_claim_ids,
        semantic_duplicate_edges,
    )
)


semantic_duplicate_component_members = {
    claim_id
    for component in (
        semantic_duplicate_components
    )
    for claim_id in (
        component
    )
}


assert (
    semantic_duplicate_component_members
    == claims_with_semantic_duplicate
)


semantic_duplicate_component_size_counts = Counter(
    len(
        component
    )
    for component in (
        semantic_duplicate_components
    )
)


semantic_duplicate_component_count = len(
    semantic_duplicate_components
)


semantic_duplicate_member_count = len(
    semantic_duplicate_component_members
)


semantic_duplicate_redundant_claim_slots = sum(
    len(
        component
    )
    - 1
    for component in (
        semantic_duplicate_components
    )
)


assert (
    semantic_duplicate_redundant_claim_slots
    == (
        semantic_duplicate_member_count
        - semantic_duplicate_component_count
    )
)


# =========================================================
# Theoretical duplicate-controlled candidate universe
#
# Important:
# This does NOT modify Registry identity.
#
# For Golden only:
# every Semantic-Duplicate connected component contributes
# at most one independent semantic slot before question
# generation.
# =========================================================

claims_outside_semantic_duplicate = (
    len(
        profile_claims
    )
    - semantic_duplicate_member_count
)


duplicate_controlled_candidate_ceiling = (
    claims_outside_semantic_duplicate
    + semantic_duplicate_component_count
)


assert (
    duplicate_controlled_candidate_ceiling
    == (
        len(
            profile_claims
        )
        - semantic_duplicate_redundant_claim_slots
    )
)


# =========================================================
# Inspect Duplicate Component authority composition
# =========================================================

semantic_duplicate_component_authority_patterns = Counter()

semantic_duplicate_mixed_authority_components = []

semantic_duplicate_single_authority_components = []


for (
    component_index,
    component,
) in enumerate(
    semantic_duplicate_components,
    start=1,
):

    authorities = sorted({
        profile_claim_by_id[
            claim_id
        ][
            "authority_classes"
        ][
            0
        ]
        for claim_id in component
    })


    authority_pattern = (
        " + ".join(
            authorities
        )
    )


    semantic_duplicate_component_authority_patterns[
        authority_pattern
    ] += 1


    component_record = {
        "component_index": (
            component_index
        ),

        "size": (
            len(
                component
            )
        ),

        "claim_ids": (
            component
        ),

        "authorities": (
            authorities
        ),
    }


    if (
        len(
            authorities
        )
        > 1
    ):

        semantic_duplicate_mixed_authority_components.append(
            component_record
        )

    else:

        semantic_duplicate_single_authority_components.append(
            component_record
        )


# =========================================================
# Duplicate Component source-type composition
# =========================================================

semantic_duplicate_component_source_patterns = Counter()

semantic_duplicate_mixed_source_type_components = []


for (
    component_index,
    component,
) in enumerate(
    semantic_duplicate_components,
    start=1,
):

    source_types_in_component = sorted({
        source_type
        for claim_id in component
        for source_type in (
            profile_claim_by_id[
                claim_id
            ][
                "source_types"
            ]
        )
    })


    pattern = (
        " + ".join(
            source_types_in_component
        )
    )


    semantic_duplicate_component_source_patterns[
        pattern
    ] += 1


    if (
        len(
            source_types_in_component
        )
        > 1
    ):

        semantic_duplicate_mixed_source_type_components.append({
            "component_index": (
                component_index
            ),

            "size": (
                len(
                    component
                )
            ),

            "source_types": (
                source_types_in_component
            ),

            "claim_ids": (
                component
            ),
        })


# =========================================================
# Relation membership combinations
# =========================================================

relation_membership_combination_counts = Counter()

relation_membership_claim_ids = defaultdict(
    list
)


for claim_id in sorted(
    all_claim_ids
):

    memberships = []


    if (
        claim_id
        in claims_with_semantic_duplicate
    ):

        memberships.append(
            "SEMANTIC_DUPLICATE"
        )


    if (
        claim_id
        in claims_with_subsumes
    ):

        memberships.append(
            "SUBSUMES"
        )


    if (
        claim_id
        in claims_with_partial_overlap
    ):

        memberships.append(
            "PARTIAL_OVERLAP"
        )


    if not memberships:

        combination = (
            "NONE"
        )

    else:

        combination = (
            " + ".join(
                memberships
            )
        )


    relation_membership_combination_counts[
        combination
    ] += 1


    relation_membership_claim_ids[
        combination
    ].append(
        claim_id
    )


assert (
    sum(
        relation_membership_combination_counts.values()
    )
    == 4006
)


assert (
    relation_membership_combination_counts[
        "NONE"
    ]
    == 2743
)


# =========================================================
# Review populations after Semantic-Duplicate control
# =========================================================

claims_needing_duplicate_representative_selection = (
    semantic_duplicate_component_members
)


claims_with_subsumes_not_semantic_duplicate = (
    claims_with_subsumes
    - claims_with_semantic_duplicate
)


claims_with_partial_overlap_not_semantic_duplicate = (
    claims_with_partial_overlap
    - claims_with_semantic_duplicate
)


claims_with_overlap_or_subsumption_after_duplicate_control = (
    (
        claims_with_subsumes
        | claims_with_partial_overlap
    )
    - claims_with_semantic_duplicate
)


# =========================================================
# All semantic-relation connected components
#
# This is diagnostic only.
# It shows how large the overall relation neighborhoods are.
# =========================================================

all_semantic_relation_edges = (
    semantic_duplicate_edges
    + subsumes_edges
    + partial_overlap_edges
)


all_relation_components = (
    connected_components_from_edges(
        all_claim_ids,
        all_semantic_relation_edges,
    )
)


all_relation_component_size_counts = Counter(
    len(
        component
    )
    for component in (
        all_relation_components
    )
)


all_relation_component_members = {
    claim_id
    for component in (
        all_relation_components
    )
    for claim_id in (
        component
    )
}


assert (
    all_relation_component_members
    == claims_with_any_graph_relation
)


largest_relation_component_size = (
    max(
        (
            len(
                component
            )
            for component in (
                all_relation_components
            )
        ),
        default=0,
    )
)


largest_semantic_duplicate_component_size = (
    max(
        (
            len(
                component
            )
            for component in (
                semantic_duplicate_components
            )
        ),
        default=0,
    )
)


# =========================================================
# Claim-Type profile after theoretical Duplicate control
#
# No representative is chosen yet, so we report:
# - claims outside duplicate components
# - duplicate components by member Claim Types
# =========================================================

claims_outside_duplicate_by_type = Counter()


for claim in (
    profile_claims
):

    if (
        claim[
            "claim_id"
        ]
        in semantic_duplicate_component_members
    ):

        continue


    for claim_type in (
        claim[
            "claim_types"
        ]
    ):

        claims_outside_duplicate_by_type[
            claim_type
        ] += 1


duplicate_component_claim_type_patterns = Counter()


for component in (
    semantic_duplicate_components
):

    types = sorted({
        claim_type
        for claim_id in component
        for claim_type in (
            profile_claim_by_id[
                claim_id
            ][
                "claim_types"
            ]
        )
    })


    duplicate_component_claim_type_patterns[
        " + ".join(
            types
        )
    ] += 1


# =========================================================
# Cost / adjudication boundary
#
# Relation adjudication is already complete in NB12i.
# What remains is Seed-selection policy and later
# Question-level independence / quality validation.
# =========================================================

semantic_relation_adjudication_already_complete = (
    True
)


new_semantic_relation_adjudication_required_now = (
    False
)


duplicate_component_representative_decisions_required = (
    semantic_duplicate_component_count
)


subsumes_pairs_for_question_independence_awareness = (
    len(
        subsumes_edges
    )
)


partial_overlap_pairs_for_question_redundancy_awareness = (
    len(
        partial_overlap_edges
    )
)


# =========================================================
# Stage Gate
# =========================================================

seed_suitability_discovery_status = (
    "PASS"
)


seed_selection_contract_readiness = (
    "READY_FOR_SEED_SELECTION_CONTRACT"
)


# =========================================================
# Result
# =========================================================

print(
    "=== NB12j SEED SUITABILITY DISCOVERY ==="
)

print()

print(
    "Canonical structurally eligible Claims:",
    len(
        profile_claims
    ),
)


print()

print(
    "=== SEMANTIC DUPLICATE GROUPS ==="
)

print(
    "SEMANTIC_DUPLICATE relations:",
    len(
        semantic_duplicate_edges
    ),
)

print(
    "Claims inside Semantic-Duplicate components:",
    semantic_duplicate_member_count,
)

print(
    "Semantic-Duplicate connected components:",
    semantic_duplicate_component_count,
)

print(
    "Component size distribution:",
    dict(
        sorted(
            semantic_duplicate_component_size_counts.items()
        )
    ),
)

print(
    "Largest Semantic-Duplicate component:",
    largest_semantic_duplicate_component_size,
)

print(
    "Redundant Claim slots if one semantic slot "
    "is kept per component:",
    semantic_duplicate_redundant_claim_slots,
)


print()

print(
    "Claims outside Semantic-Duplicate components:",
    claims_outside_semantic_duplicate,
)

print(
    "Duplicate-controlled candidate ceiling:",
    duplicate_controlled_candidate_ceiling,
)


print()

print(
    "Semantic-Duplicate authority patterns:",
    dict(
        sorted(
            semantic_duplicate_component_authority_patterns.items()
        )
    ),
)

print(
    "Mixed-authority Semantic-Duplicate components:",
    len(
        semantic_duplicate_mixed_authority_components
    ),
)

print(
    "Semantic-Duplicate source-type patterns:",
    dict(
        sorted(
            semantic_duplicate_component_source_patterns.items()
        )
    ),
)

print(
    "Mixed-source-type Semantic-Duplicate components:",
    len(
        semantic_duplicate_mixed_source_type_components
    ),
)


print()

print(
    "=== RELATION MEMBERSHIP COMBINATIONS ==="
)

for (
    combination,
    count,
) in sorted(
    relation_membership_combination_counts.items()
):

    print(
        "  ",
        combination,
        ":",
        count,
    )


print()

print(
    "Claims needing Duplicate representative selection:",
    len(
        claims_needing_duplicate_representative_selection
    ),
)

print(
    "Duplicate components requiring one representative decision:",
    duplicate_component_representative_decisions_required,
)

print(
    "Claims with SUBSUMES but no Semantic-Duplicate flag:",
    len(
        claims_with_subsumes_not_semantic_duplicate
    ),
)

print(
    "Claims with PARTIAL_OVERLAP but no Semantic-Duplicate flag:",
    len(
        claims_with_partial_overlap_not_semantic_duplicate
    ),
)

print(
    "Claims with SUBSUMES and/or PARTIAL_OVERLAP "
    "after excluding Semantic-Duplicate members:",
    len(
        claims_with_overlap_or_subsumption_after_duplicate_control
    ),
)


print()

print(
    "=== OVERALL RELATION NEIGHBORHOODS ==="
)

print(
    "Claims in any semantic relation neighborhood:",
    len(
        all_relation_component_members
    ),
)

print(
    "All-relation connected components:",
    len(
        all_relation_components
    ),
)

print(
    "All-relation component size distribution:",
    dict(
        sorted(
            all_relation_component_size_counts.items()
        )
    ),
)

print(
    "Largest all-relation component:",
    largest_relation_component_size,
)


print()

print(
    "=== ADJUDICATION BOUNDARY ==="
)

print(
    "Semantic Relation adjudication already complete:",
    semantic_relation_adjudication_already_complete,
)

print(
    "New Semantic Relation adjudication required now:",
    new_semantic_relation_adjudication_required_now,
)

print(
    "SUBSUMES pairs to respect during Question design:",
    subsumes_pairs_for_question_independence_awareness,
)

print(
    "PARTIAL_OVERLAP pairs to respect during Question design:",
    partial_overlap_pairs_for_question_redundancy_awareness,
)


print()

print(
    "Fixed final Seed target decided:",
    False,
)

print(
    "Seed representatives selected:",
    False,
)

print(
    "Question generation started:",
    False,
)

print(
    "Files modified by this Cell:",
    0,
)

print(
    "API calls performed:",
    0,
)


print()

print(
    "Next stage:",
    "SEED_SELECTION_CONTRACT",
)

print(
    "Readiness:",
    seed_selection_contract_readiness,
)


print()

print(
    "CODE CELL 09 GATE:",
    seed_suitability_discovery_status,
)
````

````python
# Code Cell 10

assert (
    seed_suitability_discovery_status
    == "PASS"
), (
    "Code Cell 09 must PASS before "
    "freezing the Seed Selection Contract."
)

assert (
    seed_selection_contract_readiness
    == "READY_FOR_SEED_SELECTION_CONTRACT"
)


# =========================================================
# Policy version
# =========================================================

GOLDEN_V2_SEED_SELECTION_POLICY_VERSION = (
    "golden_v2_seed_selection_policy_v1"
)


# =========================================================
# Frozen Representative ranking rules
# =========================================================

representative_authority_priority = {
    "CORE_PRIMARY_KNOWLEDGE": 0,
    "SUPERVISED_SECONDARY_KNOWLEDGE": 1,
}


assert (
    set(
        representative_authority_priority
    )
    == set(
        observed_authority_classes
    )
)


representative_ranking_fields = [
    {
        "field": (
            "authority_class"
        ),
        "direction": (
            "PREFER_PRIMARY"
        ),
    },

    {
        "field": (
            "evidence_count"
        ),
        "direction": (
            "DESCENDING"
        ),
    },

    {
        "field": (
            "source_document_count"
        ),
        "direction": (
            "DESCENDING"
        ),
    },

    {
        "field": (
            "source_id_count"
        ),
        "direction": (
            "DESCENDING"
        ),
    },

    {
        "field": (
            "claim_id"
        ),
        "direction": (
            "ASCENDING_DETERMINISTIC_TIE_BREAKER"
        ),
    },
]


# =========================================================
# Frozen Seed Selection Contract
# =========================================================

golden_v2_seed_selection_policy = {
    "policy_version": (
        GOLDEN_V2_SEED_SELECTION_POLICY_VERSION
    ),

    "eligibility_policy_version": (
        GOLDEN_V2_ELIGIBILITY_POLICY_VERSION
    ),

    "eligibility_policy_sha256": (
        golden_v2_eligibility_policy_sha256
    ),

    "input_population": {
        "canonical_claims": (
            len(
                profile_claims
            )
        ),

        "structurally_eligible_claims": (
            len(
                structurally_eligible_claim_ids
            )
        ),

        "semantic_duplicate_member_claims": (
            semantic_duplicate_member_count
        ),

        "semantic_duplicate_components": (
            semantic_duplicate_component_count
        ),

        "claims_outside_semantic_duplicate_components": (
            claims_outside_semantic_duplicate
        ),

        "duplicate_controlled_candidate_ceiling": (
            duplicate_controlled_candidate_ceiling
        ),
    },

    "seed_unit": (
        "SEMANTIC_SEED_SLOT"
    ),

    "slot_construction": {
        "non_duplicate_claim": (
            "ONE_PROVISIONAL_SLOT_PER_CANONICAL_CLAIM"
        ),

        "semantic_duplicate_component": (
            "ONE_PROVISIONAL_SLOT_PER_CONNECTED_COMPONENT"
        ),

        "registry_claims_merged": False,

        "registry_claims_deleted": False,

        "registry_claim_ids_regenerated": False,
    },

    "semantic_duplicate_representative_policy": {
        "selection_mode": (
            "DETERMINISTIC_RANKING"
        ),

        "ranking_fields": (
            representative_ranking_fields
        ),

        "authority_priority": (
            representative_authority_priority
        ),

        "source_type_preference": None,

        "claim_type_preference": None,

        "claim_text_length_preference": None,

        "nonrepresentative_status": (
            "SEMANTIC_REDUNDANCY_NONREPRESENTATIVE"
        ),

        "nonrepresentative_claim_is_registry_reject": False,

        "preserve_component_member_claim_ids": True,

        "preserve_component_provenance": True,
    },

    "relation_policy": {
        "SEMANTIC_DUPLICATE": (
            "ONE_PROVISIONAL_SEED_SLOT_PER_COMPONENT"
        ),

        "SUBSUMES": (
            "PRESERVE_SLOT_REQUIRE_DOWNSTREAM_"
            "QUESTION_INDEPENDENCE_REVIEW"
        ),

        "PARTIAL_OVERLAP": (
            "PRESERVE_SLOT_REQUIRE_DOWNSTREAM_"
            "QUESTION_REDUNDANCY_REVIEW"
        ),

        "automatic_subsumes_slot_rejection": False,

        "automatic_partial_overlap_slot_rejection": False,
    },

    "authority_policy": {
        "primary_preferred_within_semantic_duplicate_component": True,

        "secondary_claims_allowed_as_seed_slots": True,

        "secondary_seed_requires_attribution": True,

        "secondary_may_override_primary": False,
    },

    "question_policy": {
        "default_primary_question_count_per_final_seed": 1,

        "paraphrase_creates_new_seed": False,

        "question_variant_creates_new_seed": False,

        "multiple_questions_from_one_seed_require_distinct_evaluation_intent": True,

        "question_must_be_evidence_grounded": True,

        "question_must_be_answerable_from_attached_evidence": True,

        "question_must_be_unambiguous": True,
    },

    "negative_and_challenge_policy": {
        "negative_case_creates_new_knowledge_seed": False,

        "challenge_case_creates_new_knowledge_seed": False,

        "must_preserve_original_seed_lineage": True,
    },

    "final_seed_acceptance": {
        "provisional_slot_is_final_seed": False,

        "semantic_suitability_review_required": True,

        "independent_question_possible": True,

        "unambiguous_question_possible": True,

        "evidence_grounded_answer_possible": True,

        "external_unprovided_context_required": False,

        "relation_aware_redundancy_control_required": True,

        "fixed_final_seed_target": None,

        "derive_final_count_from_actual_review": True,
    },

    "current_state": (
        "POLICY_FROZEN_PROVISIONAL_SLOT_CONSTRUCTION_PENDING"
    ),
}


# =========================================================
# Deterministic policy hash
# =========================================================

golden_v2_seed_selection_policy_text = (
    json.dumps(
        golden_v2_seed_selection_policy,
        ensure_ascii=False,
        sort_keys=True,
        separators=(
            ",",
            ":",
        ),
    )
)


golden_v2_seed_selection_policy_sha256 = (
    hashlib.sha256(
        golden_v2_seed_selection_policy_text.encode(
            "utf-8"
        )
    ).hexdigest()
)


# =========================================================
# Contract validation
# =========================================================

assert (
    golden_v2_seed_selection_policy[
        "input_population"
    ][
        "canonical_claims"
    ]
    == 4006
)


assert (
    golden_v2_seed_selection_policy[
        "input_population"
    ][
        "duplicate_controlled_candidate_ceiling"
    ]
    == duplicate_controlled_candidate_ceiling
)


assert (
    golden_v2_seed_selection_policy[
        "slot_construction"
    ][
        "registry_claims_merged"
    ]
    is False
)


assert (
    golden_v2_seed_selection_policy[
        "slot_construction"
    ][
        "registry_claims_deleted"
    ]
    is False
)


assert (
    golden_v2_seed_selection_policy[
        "relation_policy"
    ][
        "automatic_subsumes_slot_rejection"
    ]
    is False
)


assert (
    golden_v2_seed_selection_policy[
        "relation_policy"
    ][
        "automatic_partial_overlap_slot_rejection"
    ]
    is False
)


assert (
    golden_v2_seed_selection_policy[
        "question_policy"
    ][
        "paraphrase_creates_new_seed"
    ]
    is False
)


assert (
    golden_v2_seed_selection_policy[
        "final_seed_acceptance"
    ][
        "fixed_final_seed_target"
    ]
    is None
)


# =========================================================
# Expected provisional slot count
# =========================================================

expected_nonduplicate_slots = (
    claims_outside_semantic_duplicate
)


expected_duplicate_component_slots = (
    semantic_duplicate_component_count
)


expected_total_provisional_slots = (
    expected_nonduplicate_slots
    + expected_duplicate_component_slots
)


assert (
    expected_total_provisional_slots
    == duplicate_controlled_candidate_ceiling
)


# =========================================================
# Review burden after slot construction
# =========================================================

relation_aware_claims_outside_semantic_duplicate = (
    claims_with_overlap_or_subsumption_after_duplicate_control
)


assert (
    len(
        relation_aware_claims_outside_semantic_duplicate
    )
    == 962
)


# =========================================================
# Gate
# =========================================================

seed_selection_contract_status = (
    "PASS"
)


provisional_seed_slot_readiness = (
    "READY_FOR_PROVISIONAL_SEED_SLOT_CONSTRUCTION"
)


# =========================================================
# Result
# =========================================================

print(
    "=== NB12j SEED SELECTION CONTRACT ==="
)

print()

print(
    "Canonical Claims:",
    len(
        profile_claims
    ),
)

print(
    "Claims outside Semantic-Duplicate groups:",
    expected_nonduplicate_slots,
)

print(
    "Semantic-Duplicate Components:",
    expected_duplicate_component_slots,
)

print(
    "Expected Provisional Seed Slots:",
    expected_total_provisional_slots,
)


print()

print(
    "Representative selection:",
    golden_v2_seed_selection_policy[
        "semantic_duplicate_representative_policy"
    ][
        "selection_mode"
    ],
)

print(
    "Primary preferred over Secondary:",
    True,
)

print(
    "Source Type preference:",
    None,
)

print(
    "Claim Type preference:",
    None,
)

print(
    "Text Length preference:",
    None,
)


print()

print(
    "Nonrepresentative Duplicate member rejected from Registry:",
    False,
)

print(
    "SUBSUMES automatic slot rejection:",
    False,
)

print(
    "PARTIAL_OVERLAP automatic slot rejection:",
    False,
)


print()

print(
    "Relation-aware non-duplicate Claims remaining:",
    len(
        relation_aware_claims_outside_semantic_duplicate
    ),
)

print(
    "Default primary Questions per Final Seed:",
    golden_v2_seed_selection_policy[
        "question_policy"
    ][
        "default_primary_question_count_per_final_seed"
    ],
)

print(
    "Paraphrase creates new Seed:",
    False,
)


print()

print(
    "Fixed Final Seed target:",
    golden_v2_seed_selection_policy[
        "final_seed_acceptance"
    ][
        "fixed_final_seed_target"
    ],
)

print(
    "Semantic Suitability Review required:",
    golden_v2_seed_selection_policy[
        "final_seed_acceptance"
    ][
        "semantic_suitability_review_required"
    ],
)


print()

print(
    "Seed Selection Policy version:",
    GOLDEN_V2_SEED_SELECTION_POLICY_VERSION,
)

print(
    "Seed Selection Policy SHA-256:",
    golden_v2_seed_selection_policy_sha256,
)

print(
    "Files modified by this Cell:",
    0,
)

print(
    "API calls performed:",
    0,
)


print()

print(
    "Next stage:",
    "PROVISIONAL_SEED_SLOT_CONSTRUCTION",
)

print(
    "Readiness:",
    provisional_seed_slot_readiness,
)


print()

print(
    "CODE CELL 10 GATE:",
    seed_selection_contract_status,
)
````

````python
# Code Cell 11

assert (
    seed_selection_contract_status
    == "PASS"
), (
    "Code Cell 10 must PASS before "
    "Provisional Seed Slot Construction."
)

assert (
    provisional_seed_slot_readiness
    == "READY_FOR_PROVISIONAL_SEED_SLOT_CONSTRUCTION"
)


# =========================================================
# Configuration
# =========================================================

PROVISIONAL_SEED_POOL_VERSION = (
    "golden_v2_seed_pool_v1"
)


provisional_seed_pool_path = (
    golden_v2_artifact_root
    / golden_v2_artifact_family[
        "seed_pool"
    ]
)


# =========================================================
# Representative ranking
# =========================================================

def representative_rank(
    claim_id,
):

    claim = (
        profile_claim_by_id[
            claim_id
        ]
    )


    authority_class = (
        claim[
            "authority_classes"
        ][
            0
        ]
    )


    return (
        representative_authority_priority[
            authority_class
        ],

        -claim[
            "evidence_count"
        ],

        -len(
            claim[
                "source_document_ids"
            ]
        ),

        -len(
            claim[
                "source_ids"
            ]
        ),

        claim_id,
    )


# =========================================================
# Build semantic units
# =========================================================

semantic_duplicate_member_set = set(
    semantic_duplicate_component_members
)


singleton_member_sets = [
    [
        claim_id
    ]
    for claim_id in sorted(
        all_claim_ids
        - semantic_duplicate_member_set
    )
]


duplicate_member_sets = [
    sorted(
        component
    )
    for component in (
        semantic_duplicate_components
    )
]


assert (
    len(
        singleton_member_sets
    )
    == 3705
)


assert (
    len(
        duplicate_member_sets
    )
    == 112
)


all_seed_member_sets = (
    singleton_member_sets
    + duplicate_member_sets
)


assert (
    len(
        all_seed_member_sets
    )
    == 3817
)


# =========================================================
# Stable Seed Slot IDs
# =========================================================

def make_seed_slot_id(
    member_claim_ids,
):

    payload = {
        "policy_version": (
            GOLDEN_V2_SEED_SELECTION_POLICY_VERSION
        ),

        "member_claim_ids": sorted(
            member_claim_ids
        ),
    }


    payload_text = json.dumps(
        payload,
        ensure_ascii=False,
        sort_keys=True,
        separators=(
            ",",
            ":",
        ),
    )


    digest = hashlib.sha256(
        payload_text.encode(
            "utf-8"
        )
    ).hexdigest()


    return (
        "gsv2_slot_"
        + digest[
            :24
        ]
    )


# =========================================================
# Build preliminary Slot records
# =========================================================

preliminary_seed_slots = []

claim_to_seed_slot_id = {}


for member_claim_ids in (
    all_seed_member_sets
):

    member_claim_ids = sorted(
        member_claim_ids
    )


    representative_claim_id = min(
        member_claim_ids,
        key=representative_rank,
    )


    representative_claim = (
        profile_claim_by_id[
            representative_claim_id
        ]
    )


    seed_slot_id = (
        make_seed_slot_id(
            member_claim_ids
        )
    )


    assert (
        seed_slot_id
        not in {
            record[
                "seed_slot_id"
            ]
            for record in (
                preliminary_seed_slots
            )
        }
    ), (
        "Seed Slot ID collision: "
        f"{seed_slot_id}"
    )


    for claim_id in (
        member_claim_ids
    ):

        assert (
            claim_id
            not in claim_to_seed_slot_id
        )

        claim_to_seed_slot_id[
            claim_id
        ] = seed_slot_id


    member_claims = [
        profile_claim_by_id[
            claim_id
        ]
        for claim_id in (
            member_claim_ids
        )
    ]


    member_authorities = sorted({
        claim[
            "authority_classes"
        ][
            0
        ]
        for claim in (
            member_claims
        )
    })


    member_source_types = sorted({
        source_type
        for claim in (
            member_claims
        )
        for source_type in (
            claim[
                "source_types"
            ]
        )
    })


    member_claim_types = sorted({
        claim_type
        for claim in (
            member_claims
        )
        for claim_type in (
            claim[
                "claim_types"
            ]
        )
    })


    component_evidence_ids = sorted({
        evidence_id
        for claim in (
            member_claims
        )
        for evidence_id in (
            claim[
                "evidence_ids"
            ]
        )
    })


    component_source_ids = sorted({
        source_id
        for claim in (
            member_claims
        )
        for source_id in (
            claim[
                "source_ids"
            ]
        )
    })


    component_document_ids = sorted({
        document_id
        for claim in (
            member_claims
        )
        for document_id in (
            claim[
                "source_document_ids"
            ]
        )
    })


    nonrepresentative_claim_ids = [
        claim_id
        for claim_id in (
            member_claim_ids
        )
        if (
            claim_id
            != representative_claim_id
        )
    ]


    is_semantic_duplicate_component = (
        len(
            member_claim_ids
        )
        > 1
    )


    preliminary_seed_slots.append({
        "seed_pool_version": (
            PROVISIONAL_SEED_POOL_VERSION
        ),

        "seed_selection_policy_version": (
            GOLDEN_V2_SEED_SELECTION_POLICY_VERSION
        ),

        "seed_selection_policy_sha256": (
            golden_v2_seed_selection_policy_sha256
        ),

        "seed_slot_id": (
            seed_slot_id
        ),

        "slot_kind": (
            "SEMANTIC_DUPLICATE_COMPONENT"
            if is_semantic_duplicate_component
            else "SINGLE_CANONICAL_CLAIM"
        ),

        "member_claim_ids": (
            member_claim_ids
        ),

        "member_claim_count": (
            len(
                member_claim_ids
            )
        ),

        "representative_claim_id": (
            representative_claim_id
        ),

        "nonrepresentative_claim_ids": (
            nonrepresentative_claim_ids
        ),

        "representative_selection_mode": (
            "DETERMINISTIC_RANKING"
            if is_semantic_duplicate_component
            else "SINGLETON_IDENTITY"
        ),

        "representative_claim": {
            "claim_text_fa": (
                representative_claim[
                    "claim_text_fa"
                ]
            ),

            "claim_text_sha256": (
                representative_claim[
                    "claim_text_sha256"
                ]
            ),

            "claim_types": (
                representative_claim[
                    "claim_types"
                ]
            ),

            "authority_class": (
                representative_claim[
                    "authority_classes"
                ][
                    0
                ]
            ),

            "source_types": (
                representative_claim[
                    "source_types"
                ]
            ),

            "source_ids": (
                representative_claim[
                    "source_ids"
                ]
            ),

            "source_document_ids": (
                representative_claim[
                    "source_document_ids"
                ]
            ),

            "evidence_ids": (
                representative_claim[
                    "evidence_ids"
                ]
            ),

            "evidence_count": (
                representative_claim[
                    "evidence_count"
                ]
            ),
        },

        "component_profile": {
            "authority_classes": (
                member_authorities
            ),

            "source_types": (
                member_source_types
            ),

            "claim_types": (
                member_claim_types
            ),

            "component_evidence_ids": (
                component_evidence_ids
            ),

            "component_evidence_count": (
                len(
                    component_evidence_ids
                )
            ),

            "component_source_ids": (
                component_source_ids
            ),

            "component_source_count": (
                len(
                    component_source_ids
                )
            ),

            "component_document_ids": (
                component_document_ids
            ),

            "component_document_count": (
                len(
                    component_document_ids
                )
            ),

            "mixed_authority": (
                len(
                    member_authorities
                )
                > 1
            ),

            "mixed_source_type": (
                len(
                    member_source_types
                )
                > 1
            ),
        },

        "evidence_selection_status": (
            "NOT_STARTED"
        ),

        "semantic_suitability_status": (
            "PENDING_SEMANTIC_SUITABILITY_REVIEW"
        ),

        "final_seed_selected": (
            False
        ),

        "question_generation_started": (
            False
        ),
    })


# =========================================================
# Validate total Claim preservation
# =========================================================

assert (
    len(
        preliminary_seed_slots
    )
    == 3817
)


assert (
    len(
        claim_to_seed_slot_id
    )
    == 4006
)


assert (
    set(
        claim_to_seed_slot_id
    )
    == all_claim_ids
)


all_member_claim_occurrences = sum(
    slot[
        "member_claim_count"
    ]
    for slot in (
        preliminary_seed_slots
    )
)


assert (
    all_member_claim_occurrences
    == 4006
)


nonrepresentative_member_count = sum(
    len(
        slot[
            "nonrepresentative_claim_ids"
        ]
    )
    for slot in (
        preliminary_seed_slots
    )
)


assert (
    nonrepresentative_member_count
    == semantic_duplicate_redundant_claim_slots
)


assert (
    nonrepresentative_member_count
    == 189
)


# =========================================================
# Validate Representative policy
# =========================================================

duplicate_slots = [
    slot
    for slot in (
        preliminary_seed_slots
    )
    if (
        slot[
            "slot_kind"
        ]
        == "SEMANTIC_DUPLICATE_COMPONENT"
    )
]


singleton_slots = [
    slot
    for slot in (
        preliminary_seed_slots
    )
    if (
        slot[
            "slot_kind"
        ]
        == "SINGLE_CANONICAL_CLAIM"
    )
]


assert (
    len(
        duplicate_slots
    )
    == 112
)


assert (
    len(
        singleton_slots
    )
    == 3705
)


for slot in (
    duplicate_slots
):

    expected_representative = min(
        slot[
            "member_claim_ids"
        ],
        key=representative_rank,
    )


    assert (
        slot[
            "representative_claim_id"
        ]
        == expected_representative
    )


# Mixed-authority Duplicate Components must select Primary.
mixed_authority_duplicate_slots = [
    slot
    for slot in (
        duplicate_slots
    )
    if (
        slot[
            "component_profile"
        ][
            "mixed_authority"
        ]
    )
]


assert (
    len(
        mixed_authority_duplicate_slots
    )
    == 23
)


assert all(
    slot[
        "representative_claim"
    ][
        "authority_class"
    ]
    == "CORE_PRIMARY_KNOWLEDGE"
    for slot in (
        mixed_authority_duplicate_slots
    )
)


# =========================================================
# Relation index
# =========================================================

relation_by_id = {
    relation[
        "relation_id"
    ]: relation
    for relation in (
        graph_relations
    )
}


assert (
    len(
        relation_by_id
    )
    == len(
        graph_relations
    )
)


# =========================================================
# Project Claim Relations onto Seed Slots
# =========================================================

slot_relation_profiles = {
    slot[
        "seed_slot_id"
    ]: {
        "internal_relation_ids": {
            "SEMANTIC_DUPLICATE": [],
            "SUBSUMES": [],
            "PARTIAL_OVERLAP": [],
        },

        "external_relation_ids": {
            "SEMANTIC_DUPLICATE": [],
            "SUBSUMES": [],
            "PARTIAL_OVERLAP": [],
        },

        "external_neighbor_seed_slot_ids": {
            "SEMANTIC_DUPLICATE": set(),
            "SUBSUMES": set(),
            "PARTIAL_OVERLAP": set(),
        },
    }
    for slot in (
        preliminary_seed_slots
    )
}


internal_projected_relation_counts = Counter()

external_projected_relation_counts = Counter()


for relation in (
    graph_relations
):

    relation_id = (
        relation[
            "relation_id"
        ]
    )

    relation_type = (
        relation[
            "relation_type"
        ]
    )

    claim_id_a = (
        relation[
            "claim_id_a"
        ]
    )

    claim_id_b = (
        relation[
            "claim_id_b"
        ]
    )


    slot_id_a = (
        claim_to_seed_slot_id[
            claim_id_a
        ]
    )

    slot_id_b = (
        claim_to_seed_slot_id[
            claim_id_b
        ]
    )


    if (
        slot_id_a
        == slot_id_b
    ):

        internal_projected_relation_counts[
            relation_type
        ] += 1


        slot_relation_profiles[
            slot_id_a
        ][
            "internal_relation_ids"
        ][
            relation_type
        ].append(
            relation_id
        )


    else:

        external_projected_relation_counts[
            relation_type
        ] += 1


        for (
            source_slot_id,
            neighbor_slot_id,
        ) in [
            (
                slot_id_a,
                slot_id_b,
            ),
            (
                slot_id_b,
                slot_id_a,
            ),
        ]:

            slot_relation_profiles[
                source_slot_id
            ][
                "external_relation_ids"
            ][
                relation_type
            ].append(
                relation_id
            )


            slot_relation_profiles[
                source_slot_id
            ][
                "external_neighbor_seed_slot_ids"
            ][
                relation_type
            ].add(
                neighbor_slot_id
            )


# Every Semantic-Duplicate relation must collapse inside
# its own Semantic Duplicate component.
assert (
    external_projected_relation_counts[
        "SEMANTIC_DUPLICATE"
    ]
    == 0
)


assert (
    internal_projected_relation_counts[
        "SEMANTIC_DUPLICATE"
    ]
    == len(
        semantic_duplicate_edges
    )
)


assert (
    internal_projected_relation_counts[
        "SEMANTIC_DUPLICATE"
    ]
    == 233
)


# =========================================================
# Finalize Seed Slot records with projected relations
# =========================================================

provisional_seed_slot_records = []


for slot in (
    preliminary_seed_slots
):

    seed_slot_id = (
        slot[
            "seed_slot_id"
        ]
    )


    relation_profile = (
        slot_relation_profiles[
            seed_slot_id
        ]
    )


    external_subsumes_count = len(
        relation_profile[
            "external_relation_ids"
        ][
            "SUBSUMES"
        ]
    )


    external_partial_overlap_count = len(
        relation_profile[
            "external_relation_ids"
        ][
            "PARTIAL_OVERLAP"
        ]
    )


    review_flags = []


    if (
        slot[
            "slot_kind"
        ]
        == "SEMANTIC_DUPLICATE_COMPONENT"
    ):

        review_flags.append(
            "SEMANTIC_DUPLICATE_COMPONENT"
        )


    if (
        slot[
            "component_profile"
        ][
            "mixed_authority"
        ]
    ):

        review_flags.append(
            "MIXED_AUTHORITY_COMPONENT"
        )


    if (
        slot[
            "component_profile"
        ][
            "mixed_source_type"
        ]
    ):

        review_flags.append(
            "MIXED_SOURCE_TYPE_COMPONENT"
        )


    if (
        slot[
            "representative_claim"
        ][
            "authority_class"
        ]
        == "SUPERVISED_SECONDARY_KNOWLEDGE"
    ):

        review_flags.append(
            "ATTRIBUTION_REQUIRED"
        )


    if (
        external_subsumes_count
        > 0
    ):

        review_flags.append(
            "SUBSUMPTION_REVIEW"
        )


    if (
        external_partial_overlap_count
        > 0
    ):

        review_flags.append(
            "PARTIAL_OVERLAP_REVIEW"
        )


    if (
        len(
            slot[
                "representative_claim"
            ][
                "claim_text_fa"
            ]
        )
        <= 40
    ):

        review_flags.append(
            "SHORT_TEXT_DIAGNOSTIC"
        )


    relation_projection = {
        "internal_relation_ids": {
            relation_type: sorted(
                relation_ids
            )
            for (
                relation_type,
                relation_ids,
            ) in (
                relation_profile[
                    "internal_relation_ids"
                ].items()
            )
        },

        "external_relation_ids": {
            relation_type: sorted(
                relation_ids
            )
            for (
                relation_type,
                relation_ids,
            ) in (
                relation_profile[
                    "external_relation_ids"
                ].items()
            )
        },

        "external_neighbor_seed_slot_ids": {
            relation_type: sorted(
                neighbor_ids
            )
            for (
                relation_type,
                neighbor_ids,
            ) in (
                relation_profile[
                    "external_neighbor_seed_slot_ids"
                ].items()
            )
        },

        "external_subsumes_relation_count": (
            external_subsumes_count
        ),

        "external_partial_overlap_relation_count": (
            external_partial_overlap_count
        ),

        "relation_aware_review_required": (
            (
                external_subsumes_count
                > 0
            )
            or (
                external_partial_overlap_count
                > 0
            )
        ),
    }


    final_record = dict(
        slot
    )


    final_record[
        "relation_projection"
    ] = relation_projection


    final_record[
        "review_flags"
    ] = sorted(
        review_flags
    )


    final_record[
        "review_flag_count"
    ] = len(
        review_flags
    )


    final_record[
        "seed_pool_status"
    ] = (
        "PROVISIONAL_SEED_SLOT"
    )


    provisional_seed_slot_records.append(
        final_record
    )


# =========================================================
# Deterministic ordering
# =========================================================

provisional_seed_slot_records = sorted(
    provisional_seed_slot_records,
    key=lambda record: (
        record[
            "representative_claim_id"
        ],
        record[
            "seed_slot_id"
        ],
    ),
)


# =========================================================
# Slot-level validation
# =========================================================

assert (
    len(
        provisional_seed_slot_records
    )
    == 3817
)


seed_slot_ids = [
    record[
        "seed_slot_id"
    ]
    for record in (
        provisional_seed_slot_records
    )
]


assert (
    len(
        seed_slot_ids
    )
    == len(
        set(
            seed_slot_ids
        )
    )
)


representative_claim_ids = [
    record[
        "representative_claim_id"
    ]
    for record in (
        provisional_seed_slot_records
    )
]


assert (
    len(
        representative_claim_ids
    )
    == len(
        set(
            representative_claim_ids
        )
    )
)


assert all(
    record[
        "final_seed_selected"
    ]
    is False
    for record in (
        provisional_seed_slot_records
    )
)


assert all(
    record[
        "semantic_suitability_status"
    ]
    == "PENDING_SEMANTIC_SUITABILITY_REVIEW"
    for record in (
        provisional_seed_slot_records
    )
)


assert all(
    record[
        "question_generation_started"
    ]
    is False
    for record in (
        provisional_seed_slot_records
    )
)


# =========================================================
# Slot-level diagnostics
# =========================================================

representative_authority_counts = Counter(
    record[
        "representative_claim"
    ][
        "authority_class"
    ]
    for record in (
        provisional_seed_slot_records
    )
)


representative_source_type_counts = Counter(
    source_type
    for record in (
        provisional_seed_slot_records
    )
    for source_type in (
        record[
            "representative_claim"
        ][
            "source_types"
        ]
    )
)


representative_claim_type_counts = Counter(
    claim_type
    for record in (
        provisional_seed_slot_records
    )
    for claim_type in (
        record[
            "representative_claim"
        ][
            "claim_types"
        ]
    )
)


slot_review_flag_counts = Counter(
    flag
    for record in (
        provisional_seed_slot_records
    )
    for flag in (
        record[
            "review_flags"
        ]
    )
)


relation_aware_seed_slots = [
    record[
        "seed_slot_id"
    ]
    for record in (
        provisional_seed_slot_records
    )
    if (
        record[
            "relation_projection"
        ][
            "relation_aware_review_required"
        ]
    )
]


relation_isolated_seed_slots = [
    record[
        "seed_slot_id"
    ]
    for record in (
        provisional_seed_slot_records
    )
    if not (
        record[
            "relation_projection"
        ][
            "relation_aware_review_required"
        ]
    )
]


# =========================================================
# Preserve total Evidence lineage
# =========================================================

component_evidence_union = {
    evidence_id
    for record in (
        provisional_seed_slot_records
    )
    for evidence_id in (
        record[
            "component_profile"
        ][
            "component_evidence_ids"
        ]
    )
}


assert (
    component_evidence_union
    == registry_evidence_ids
)


assert (
    len(
        component_evidence_union
    )
    == 4125
)


# =========================================================
# Serialize provisional Seed Pool
# =========================================================

provisional_seed_pool_text = "".join(
    json.dumps(
        record,
        ensure_ascii=False,
        sort_keys=True,
    )
    + "\n"
    for record in (
        provisional_seed_slot_records
    )
)


provisional_seed_pool_sha256 = (
    hashlib.sha256(
        provisional_seed_pool_text.encode(
            "utf-8"
        )
    ).hexdigest()
)


provisional_seed_pool_write_status = (
    publish_frozen_text(
        provisional_seed_pool_path,
        provisional_seed_pool_text,
    )
)


# =========================================================
# Independent disk validation
# =========================================================

disk_provisional_seed_pool = (
    load_jsonl_records(
        provisional_seed_pool_path
    )
)


assert (
    disk_provisional_seed_pool
    == provisional_seed_slot_records
)


assert (
    sha256_file(
        provisional_seed_pool_path
    )
    == provisional_seed_pool_sha256
)


assert (
    len(
        disk_provisional_seed_pool
    )
    == 3817
)


# =========================================================
# Gate
# =========================================================

provisional_seed_slot_status = (
    "PASS"
)


semantic_seed_suitability_review_readiness = (
    "READY_FOR_SEMANTIC_SEED_SUITABILITY_REVIEW_DESIGN"
)


# =========================================================
# Result
# =========================================================

print(
    "=== NB12j PROVISIONAL SEED SLOT CONSTRUCTION ==="
)

print()

print(
    "Provisional Seed Slots:",
    len(
        provisional_seed_slot_records
    ),
)

print(
    "Singleton Slots:",
    len(
        singleton_slots
    ),
)

print(
    "Semantic-Duplicate Component Slots:",
    len(
        duplicate_slots
    ),
)

print(
    "Canonical Claims represented:",
    all_member_claim_occurrences,
)

print(
    "Nonrepresentative Duplicate members preserved:",
    nonrepresentative_member_count,
)


print()

print(
    "Mixed-authority Duplicate Slots:",
    len(
        mixed_authority_duplicate_slots
    ),
)

print(
    "Mixed-authority Slots with Primary representative:",
    sum(
        1
        for slot in (
            mixed_authority_duplicate_slots
        )
        if (
            slot[
                "representative_claim"
            ][
                "authority_class"
            ]
            == "CORE_PRIMARY_KNOWLEDGE"
        )
    ),
)


print()

print(
    "Representative Authority counts:",
    dict(
        sorted(
            representative_authority_counts.items()
        )
    ),
)

print(
    "Representative Source-Type counts:",
    dict(
        sorted(
            representative_source_type_counts.items()
        )
    ),
)

print(
    "Representative Claim-Type counts:",
    dict(
        sorted(
            representative_claim_type_counts.items()
        )
    ),
)


print()

print(
    "=== RELATION PROJECTION ==="
)

print(
    "Internal projected Relations:",
    dict(
        sorted(
            internal_projected_relation_counts.items()
        )
    ),
)

print(
    "External projected Relations:",
    dict(
        sorted(
            external_projected_relation_counts.items()
        )
    ),
)

print(
    "External Semantic-Duplicate Relations:",
    external_projected_relation_counts[
        "SEMANTIC_DUPLICATE"
    ],
)

print(
    "Seed Slots requiring SUBSUMES/PARTIAL review:",
    len(
        relation_aware_seed_slots
    ),
)

print(
    "Seed Slots without external SUBSUMES/PARTIAL review:",
    len(
        relation_isolated_seed_slots
    ),
)


print()

print(
    "Seed Slot review flag counts:",
    dict(
        sorted(
            slot_review_flag_counts.items()
        )
    ),
)


print()

print(
    "Unique Registry Evidence IDs preserved through components:",
    len(
        component_evidence_union
    ),
)


print()

print(
    "Final Seed Slots selected:",
    sum(
        1
        for record in (
            provisional_seed_slot_records
        )
        if (
            record[
                "final_seed_selected"
            ]
        )
    ),
)

print(
    "Question generation started:",
    any(
        record[
            "question_generation_started"
        ]
        for record in (
            provisional_seed_slot_records
        )
    ),
)


print()

print(
    "Provisional Seed Pool:",
    provisional_seed_pool_path.relative_to(
        project_root
    ),
)

print(
    "Provisional Seed Pool SHA-256:",
    provisional_seed_pool_sha256,
)

print(
    "Write status:",
    provisional_seed_pool_write_status,
)


print()

print(
    "API calls performed:",
    0,
)

print(
    "Next stage:",
    "SEMANTIC_SEED_SUITABILITY_REVIEW_DESIGN",
)

print(
    "Readiness:",
    semantic_seed_suitability_review_readiness,
)


print()

print(
    "CODE CELL 11 GATE:",
    provisional_seed_slot_status,
)
````

<h3>Verdictهای مجاز</h3>

<ul>
<li><code>ACCEPT_SEED</code>: Slot قابلیت ساخت سؤال مستقل و Evidence-Grounded دارد.</li>
<li><code>HOLD_REPAIR</code>: Claim معتبر است اما Scope، Evidence Selection یا Question Framing نیاز به Repair دارد.</li>
<li><code>HOLD_RELATION_REVIEW</code>: استقلال Seed نسبت به SUBSUMES یا PARTIAL_OVERLAP نیاز به بررسی بیشتر دارد.</li>

Registry و Golden دو لایه مستقل با اهداف متفاوت هستند.
</p>

<h3>Artifact Lifecycle</h3>

<p>

Semantic Judgments و Reviewed Seed Pool در Artifactهای versioned جدید نوشته می‌شوند.
برای اجرای قابل Resume نیز Cache کاری جداگانه استفاده خواهد شد و Release Artifact

````python
# Code Cell 12

assert (
    provisional_seed_slot_status
    == "PASS"
), (
    "Code Cell 11 must PASS before "
    "Semantic Seed Suitability Review design."
)

assert (
    semantic_seed_suitability_review_readiness
    == "READY_FOR_SEMANTIC_SEED_SUITABILITY_REVIEW_DESIGN"
)


# =========================================================
# Configuration
# =========================================================

SEMANTIC_REVIEW_CONTRACT_VERSION = (
    "golden_v2_semantic_seed_suitability_contract_v1"
)


semantic_review_contract_path = (
    golden_v2_artifact_root
    / "golden_v2_semantic_seed_suitability_contract_v1.json"
)


semantic_judgments_path = (
    golden_v2_artifact_root
    / "golden_v2_seed_suitability_judgments_v1.jsonl"
)


reviewed_seed_pool_path = (
    golden_v2_artifact_root
    / "golden_v2_seed_pool_reviewed_v1.jsonl"
)


semantic_review_report_path = (
    golden_v2_artifact_root
    / "golden_v2_seed_suitability_report_v1.json"
)


semantic_review_work_root = (
    golden_v2_artifact_root
    / "work"
    / "seed_suitability_judgments_v1"
)


# =========================================================
# Review scope
# =========================================================

semantic_review_population = (
    len(
        provisional_seed_slot_records
    )
)


assert (
    semantic_review_population
    == 3817
)


full_population_review_required = (
    True
)


pilot_required_before_full_review = (
    True
)


# =========================================================
# Judgment taxonomy
# =========================================================

semantic_review_verdicts = {
    "ACCEPT_SEED": (
        "The slot can support at least one independent, "
        "unambiguous, evidence-grounded benchmark question."
    ),

    "HOLD_REPAIR": (
        "The underlying knowledge is usable, but scope, "
        "evidence choice, or question framing requires repair "
        "before final seed acceptance."
    ),

    "HOLD_RELATION_REVIEW": (
        "The slot may be useful, but independence from a "
        "SUBSUMES or PARTIAL_OVERLAP neighbor is not yet "
        "sufficiently established."
    ),

    "REJECT_AS_GOLDEN_SEED": (
        "The Registry Claim remains valid, but this slot is "
        "not suitable as an independent Golden benchmark seed."
    ),
}


semantic_review_reason_codes = {
    "PASS_STANDALONE": (
        "Standalone benchmark question is possible."
    ),

    "PASS_EVIDENCE_GROUNDED": (
        "Answer can be grounded in attached evidence."
    ),

    "PASS_UNAMBIGUOUS": (
        "Question scope can be made sufficiently unambiguous."
    ),

    "PASS_INDEPENDENT": (
        "Knowledge target is sufficiently independent."
    ),

    "ATTRIBUTION_REQUIRED": (
        "Secondary knowledge must preserve attribution."
    ),

    "INSUFFICIENT_STANDALONE_CONTEXT": (
        "Claim depends on context not contained in the "
        "claim/evidence packet."
    ),

    "QUESTION_SCOPE_AMBIGUOUS": (
        "A stable benchmark question cannot yet be framed "
        "without ambiguity."
    ),

    "EVIDENCE_SCOPE_MISMATCH": (
        "Attached evidence does not cleanly support the "
        "full intended question scope."
    ),

    "SUBSUMPTION_DEPENDENCY": (
        "Seed independence is uncertain because of a "
        "SUBSUMES relationship."
    ),

    "PARTIAL_OVERLAP_DEPENDENCY": (
        "Seed independence is uncertain because of substantial "
        "PARTIAL_OVERLAP."
    ),

    "TOO_CONTEXT_DEPENDENT": (
        "The claim is valid but too dependent on surrounding "
        "context for a robust standalone benchmark item."
    ),

    "NOT_BENCHMARK_WORTHY_AS_INDEPENDENT_SEED": (
        "The claim is valid knowledge but is not useful as an "
        "independent evaluation target."
    ),
}


# =========================================================
# Required Judge dimensions
# =========================================================

semantic_review_dimensions = {
    "standalone_question_possible": (
        "BOOLEAN"
    ),

    "answerable_from_attached_evidence": (
        "BOOLEAN"
    ),

    "question_scope_unambiguous": (
        "BOOLEAN"
    ),

    "independent_evaluation_target": (
        "BOOLEAN"
    ),

    "attribution_preservable": (
        "BOOLEAN"
    ),

    "requires_external_unprovided_context": (
        "BOOLEAN"
    ),
}


# =========================================================
# Risk context
#
# Risk does NOT decide verdict.
# It only enriches the review packet.
# =========================================================

semantic_review_risk_flags = {
    "SEMANTIC_DUPLICATE_COMPONENT",
    "MIXED_AUTHORITY_COMPONENT",
    "MIXED_SOURCE_TYPE_COMPONENT",
    "ATTRIBUTION_REQUIRED",
    "SUBSUMPTION_REVIEW",
    "PARTIAL_OVERLAP_REVIEW",
    "SHORT_TEXT_DIAGNOSTIC",
}


observed_seed_review_flags = {
    flag
    for record in (
        provisional_seed_slot_records
    )
    for flag in (
        record[
            "review_flags"
        ]
    )
}


assert (
    observed_seed_review_flags
    <= semantic_review_risk_flags
)


# =========================================================
# Review policy
# =========================================================

semantic_review_policy = {
    "contract_version": (
        SEMANTIC_REVIEW_CONTRACT_VERSION
    ),

    "upstream_contract": {
        "path": str(
            golden_v2_contract_path.relative_to(
                project_root
            )
        ),

        "sha256": (
            golden_v2_contract_sha256
        ),
    },

    "upstream_provisional_seed_pool": {
        "path": str(
            provisional_seed_pool_path.relative_to(
                project_root
            )
        ),

        "sha256": (
            provisional_seed_pool_sha256
        ),

        "records": (
            semantic_review_population
        ),
    },

    "reason_for_additive_contract": (
        "The provisional Seed Pool is already frozen and "
        "must not be destructively overwritten. Semantic "
        "judgments and the reviewed Seed Pool therefore use "
        "new versioned artifacts."
    ),

    "review_scope": {
        "population": (
            semantic_review_population
        ),

        "full_population_review_required": (
            full_population_review_required
        ),

        "pilot_required_before_full_review": (
            pilot_required_before_full_review
        ),

        "pilot_selection_mode": (
            "DETERMINISTIC_OBSERVED_STRATA_COVERAGE"
        ),

        "pilot_fixed_target": None,

        "skip_low_risk_slots_without_semantic_judgment": False,
    },

    "judge_task": (
        "Determine whether each Provisional Semantic Seed Slot "
        "can serve as an independent, unambiguous, evidence-"
        "grounded Golden benchmark seed."
    ),

    "allowed_verdicts": (
        semantic_review_verdicts
    ),

    "allowed_reason_codes": (
        semantic_review_reason_codes
    ),

    "required_dimensions": (
        semantic_review_dimensions
    ),

    "relation_context_policy": {
        "semantic_duplicate_component_members_in_packet": True,

        "subsumes_neighbor_context_in_packet": True,

        "partial_overlap_neighbor_context_in_packet": True,

        "relation_context_may_auto_reject": False,

        "relation_context_informs_independence_judgment": True,
    },

    "evidence_policy": {
        "representative_evidence_required": True,

        "exact_quote_required": True,

        "locator_required": True,

        "source_id_required": True,

        "document_id_required": True,

        "secondary_attribution_required": True,

        "evidence_may_be_fabricated": False,
    },

    "registry_policy": {
        "golden_rejection_invalidates_registry_claim": False,

        "golden_rejection_deletes_registry_claim": False,

        "registry_artifacts_mutable": False,
    },

    "execution_policy": {
        "pilot_before_full_run": True,

        "full_review_after_pilot_pass": True,

        "resumable_cache_required": True,

        "one_frozen_final_judgment_per_seed_slot": True,

        "destructive_overwrite": False,

        "retry_failed_calls_allowed": True,
    },

    "release_artifacts": {
        "judgments": str(
            semantic_judgments_path.relative_to(
                project_root
            )
        ),

        "reviewed_seed_pool": str(
            reviewed_seed_pool_path.relative_to(
                project_root
            )
        ),

        "review_report": str(
            semantic_review_report_path.relative_to(
                project_root
            )
        ),
    },

    "working_cache_namespace": str(
        semantic_review_work_root.relative_to(
            project_root
        )
    ),

    "current_state": (
        "CONTRACT_FROZEN_PENDING_REVIEW_PACKET_AND_PILOT_DESIGN"
    ),
}


# =========================================================
# Deterministic Contract serialization
# =========================================================

semantic_review_policy_text = (
    json.dumps(
        semantic_review_policy,
        ensure_ascii=False,
        indent=2,
        sort_keys=True,
    )
    + "\n"
)


semantic_review_policy_sha256 = (
    hashlib.sha256(
        semantic_review_policy_text.encode(
            "utf-8"
        )
    ).hexdigest()
)


semantic_review_contract_write_status = (
    publish_frozen_text(
        semantic_review_contract_path,
        semantic_review_policy_text,
    )
)


# =========================================================
# Independent disk validation
# =========================================================

disk_semantic_review_policy = (
    load_json_file(
        semantic_review_contract_path
    )
)


assert (
    disk_semantic_review_policy
    == semantic_review_policy
)


assert (
    sha256_file(
        semantic_review_contract_path
    )
    == semantic_review_policy_sha256
)


assert (
    disk_semantic_review_policy[
        "review_scope"
    ][
        "population"
    ]
    == 3817
)


assert (
    disk_semantic_review_policy[
        "review_scope"
    ][
        "full_population_review_required"
    ]
    is True
)


assert (
    disk_semantic_review_policy[
        "review_scope"
    ][
        "skip_low_risk_slots_without_semantic_judgment"
    ]
    is False
)


assert (
    disk_semantic_review_policy[
        "registry_policy"
    ][
        "registry_artifacts_mutable"
    ]
    is False
)


# =========================================================
# Important lifecycle checks
# =========================================================

assert (
    semantic_judgments_path
    != provisional_seed_pool_path
)


assert (
    reviewed_seed_pool_path
    != provisional_seed_pool_path
)


assert (
    semantic_review_report_path
    != provisional_seed_pool_path
)


# Do not create work cache yet.
assert (
    not semantic_review_work_root.exists()
), (
    "Semantic review work cache already exists before "
    "pilot design. Inspect before proceeding."
)


# =========================================================
# Gate
# =========================================================

semantic_review_contract_status = (
    "PASS"
)


semantic_review_packet_design_readiness = (
    "READY_FOR_REVIEW_PACKET_AND_STRATIFIED_PILOT_DESIGN"
)


# =========================================================
# Result
# =========================================================

print(
    "=== NB12j SEMANTIC SEED SUITABILITY REVIEW CONTRACT ==="
)

print()

print(
    "Provisional Seed Slots:",
    semantic_review_population,
)

print(
    "Full-population Semantic Review required:",
    full_population_review_required,
)

print(
    "Low-risk Slots may bypass Semantic Judgment:",
    False,
)

print(
    "Pilot required before Full Review:",
    pilot_required_before_full_review,
)

print(
    "Pilot fixed target:",
    semantic_review_policy[
        "review_scope"
    ][
        "pilot_fixed_target"
    ],
)


print()

print(
    "Allowed verdicts:"
)

for verdict in (
    semantic_review_verdicts
):

    print(
        "  -",
        verdict,
    )


print()

print(
    "Observed review-risk flags:",
    sorted(
        observed_seed_review_flags
    ),
)


print()

print(
    "Provisional Seed Pool overwritten:",
    False,
)

print(
    "Registry artifacts mutable:",
    False,
)


print()

print(
    "Judgment artifact:",
    semantic_judgments_path.relative_to(
        project_root
    ),
)

print(
    "Reviewed Seed Pool:",
    reviewed_seed_pool_path.relative_to(
        project_root
    ),
)

print(
    "Review Report:",
    semantic_review_report_path.relative_to(
        project_root
    ),
)

print(
    "Working cache namespace:",
    semantic_review_work_root.relative_to(
        project_root
    ),
)


print()

print(
    "Semantic Review Contract:",
    semantic_review_contract_path.relative_to(
        project_root
    ),
)

print(
    "Semantic Review Contract SHA-256:",
    semantic_review_policy_sha256,
)

print(
    "Write status:",
    semantic_review_contract_write_status,
)


print()

print(
    "API calls performed:",
    0,
)

print(
    "Next stage:",
    "REVIEW_PACKET_AND_STRATIFIED_PILOT_DESIGN",
)

print(
    "Readiness:",
    semantic_review_packet_design_readiness,
)


print()

print(
    "CODE CELL 12 GATE:",
    semantic_review_contract_status,
)
````

در داخل هر Stratum، Slot دارای Relation Context بیشتر، Component بزرگ‌تر
و Evidence بیشتر برای Pilot اولویت دارد تا Prompt روی نمونه‌های چالش‌برانگیزتر

````python
# Code Cell 13

assert (
    semantic_review_contract_status
    == "PASS"
), (
    "Code Cell 12 must PASS before "
    "Review Packet and Pilot design."
)

assert (
    semantic_review_packet_design_readiness
    == "READY_FOR_REVIEW_PACKET_AND_STRATIFIED_PILOT_DESIGN"
)


# =========================================================
# Configuration
# =========================================================

REVIEW_PACKET_VERSION = (
    "golden_v2_seed_suitability_review_packet_v1"
)


PILOT_SELECTION_VERSION = (
    "golden_v2_seed_suitability_pilot_selection_v1"
)


# =========================================================
# Indexes
# =========================================================

seed_slot_by_id = {
    record[
        "seed_slot_id"
    ]: record
    for record in (
        provisional_seed_slot_records
    )
}


assert (
    len(
        seed_slot_by_id
    )
    == 3817
)


# =========================================================
# Compact Evidence packet
# =========================================================

def make_evidence_packet(
    evidence_id,
):

    evidence = (
        manifest_evidence_by_id[
            evidence_id
        ]
    )


    return {
        "evidence_id": (
            evidence_id
        ),

        "evidence_quote_exact": (
            evidence[
                "evidence_quote_exact"
            ]
        ),

        "locator": (
            evidence[
                "locator"
            ]
        ),

        "source_id": (
            evidence[
                "source_id"
            ]
        ),

        "document_id": (
            evidence[
                "document_id"
            ]
        ),

        "attribution_snapshot": (
            evidence[
                "attribution_snapshot"
            ]
        ),

        "source_type": (
            evidence.get(
                "source_type"
            )
        ),

        "authority_class": (
            evidence.get(
                "authority_class"
            )
        ),

        "claim_discovery_role": (
            evidence.get(
                "claim_discovery_role"
            )
        ),
    }


# =========================================================
# Relation neighbor lookup
# =========================================================

external_relation_context_by_slot = {
    seed_slot_id: []
    for seed_slot_id in (
        seed_slot_by_id
    )
}


for relation in (
    graph_relations
):

    relation_id = (
        relation[
            "relation_id"
        ]
    )

    relation_type = (
        relation[
            "relation_type"
        ]
    )

    claim_id_a = (
        relation[
            "claim_id_a"
        ]
    )

    claim_id_b = (
        relation[
            "claim_id_b"
        ]
    )


    slot_id_a = (
        claim_to_seed_slot_id[
            claim_id_a
        ]
    )

    slot_id_b = (
        claim_to_seed_slot_id[
            claim_id_b
        ]
    )


    # Internal Semantic-Duplicate relations are already
    # represented by Component membership.
    if (
        slot_id_a
        == slot_id_b
    ):

        continue


    # Semantic Duplicate must never remain external.
    assert (
        relation_type
        != "SEMANTIC_DUPLICATE"
    )


    for (
        current_slot_id,
        current_claim_id,
        neighbor_slot_id,
        neighbor_claim_id,
    ) in [
        (
            slot_id_a,
            claim_id_a,
            slot_id_b,
            claim_id_b,
        ),
        (
            slot_id_b,
            claim_id_b,
            slot_id_a,
            claim_id_a,
        ),
    ]:

        neighbor_slot = (
            seed_slot_by_id[
                neighbor_slot_id
            ]
        )


        external_relation_context_by_slot[
            current_slot_id
        ].append({
            "relation_id": (
                relation_id
            ),

            "relation_type": (
                relation_type
            ),

            "claim_id_a": (
                claim_id_a
            ),

            "claim_id_b": (
                claim_id_b
            ),

            "current_related_claim_id": (
                current_claim_id
            ),

            "neighbor_related_claim_id": (
                neighbor_claim_id
            ),

            "neighbor_seed_slot_id": (
                neighbor_slot_id
            ),

            "neighbor_representative_claim_id": (
                neighbor_slot[
                    "representative_claim_id"
                ]
            ),

            "neighbor_representative_claim_text_fa": (
                neighbor_slot[
                    "representative_claim"
                ][
                    "claim_text_fa"
                ]
            ),
        })


for seed_slot_id in (
    external_relation_context_by_slot
):

    external_relation_context_by_slot[
        seed_slot_id
    ] = sorted(
        external_relation_context_by_slot[
            seed_slot_id
        ],
        key=lambda item: (
            item[
                "relation_type"
            ],
            item[
                "relation_id"
            ],
            item[
                "neighbor_seed_slot_id"
            ],
        ),
    )


# =========================================================
# Build Review Packets
# =========================================================

review_packets_by_slot_id = {}


for slot in (
    provisional_seed_slot_records
):

    seed_slot_id = (
        slot[
            "seed_slot_id"
        ]
    )


    representative_claim_id = (
        slot[
            "representative_claim_id"
        ]
    )


    representative_claim = (
        profile_claim_by_id[
            representative_claim_id
        ]
    )


    representative_evidence_packets = [
        make_evidence_packet(
            evidence_id
        )
        for evidence_id in (
            representative_claim[
                "evidence_ids"
            ]
        )
    ]


    component_member_claims = [
        {
            "claim_id": (
                claim_id
            ),

            "claim_text_fa": (
                profile_claim_by_id[
                    claim_id
                ][
                    "claim_text_fa"
                ]
            ),

            "claim_types": (
                profile_claim_by_id[
                    claim_id
                ][
                    "claim_types"
                ]
            ),

            "authority_class": (
                profile_claim_by_id[
                    claim_id
                ][
                    "authority_classes"
                ][
                    0
                ]
            ),

            "source_types": (
                profile_claim_by_id[
                    claim_id
                ][
                    "source_types"
                ]
            ),

            "evidence_ids": (
                profile_claim_by_id[
                    claim_id
                ][
                    "evidence_ids"
                ]
            ),
        }
        for claim_id in (
            slot[
                "member_claim_ids"
            ]
        )
    ]


    review_packet = {
        "review_packet_version": (
            REVIEW_PACKET_VERSION
        ),

        "semantic_review_contract_version": (
            SEMANTIC_REVIEW_CONTRACT_VERSION
        ),

        "semantic_review_contract_sha256": (
            semantic_review_policy_sha256
        ),

        "seed_slot_id": (
            seed_slot_id
        ),

        "slot_kind": (
            slot[
                "slot_kind"
            ]
        ),

        "representative_claim": {
            "claim_id": (
                representative_claim_id
            ),

            "claim_text_fa": (
                representative_claim[
                    "claim_text_fa"
                ]
            ),

            "claim_types": (
                representative_claim[
                    "claim_types"
                ]
            ),

            "authority_class": (
                representative_claim[
                    "authority_classes"
                ][
                    0
                ]
            ),

            "claim_discovery_roles": (
                representative_claim[
                    "claim_discovery_roles"
                ]
            ),

            "source_types": (
                representative_claim[
                    "source_types"
                ]
            ),

            "source_ids": (
                representative_claim[
                    "source_ids"
                ]
            ),

            "source_document_ids": (
                representative_claim[
                    "source_document_ids"
                ]
            ),
        },

        "representative_evidence": (
            representative_evidence_packets
        ),

        "component_context": {
            "member_claim_count": (
                slot[
                    "member_claim_count"
                ]
            ),

            "member_claims": (
                component_member_claims
            ),

            "mixed_authority": (
                slot[
                    "component_profile"
                ][
                    "mixed_authority"
                ]
            ),

            "mixed_source_type": (
                slot[
                    "component_profile"
                ][
                    "mixed_source_type"
                ]
            ),
        },

        "relation_context": (
            external_relation_context_by_slot[
                seed_slot_id
            ]
        ),

        "review_flags": (
            slot[
                "review_flags"
            ]
        ),

        "required_judgment_dimensions": (
            semantic_review_dimensions
        ),

        "allowed_verdicts": sorted(
            semantic_review_verdicts
        ),

        "allowed_reason_codes": sorted(
            semantic_review_reason_codes
        ),
    }


    packet_text = json.dumps(
        review_packet,
        ensure_ascii=False,
        sort_keys=True,
        separators=(
            ",",
            ":",
        ),
    )


    review_packet[
        "review_packet_sha256"
    ] = hashlib.sha256(
        packet_text.encode(
            "utf-8"
        )
    ).hexdigest()


    assert (
        seed_slot_id
        not in review_packets_by_slot_id
    )


    review_packets_by_slot_id[
        seed_slot_id
    ] = review_packet


# =========================================================
# Review Packet validation
# =========================================================

assert (
    len(
        review_packets_by_slot_id
    )
    == 3817
)


assert (
    set(
        review_packets_by_slot_id
    )
    == set(
        seed_slot_by_id
    )
)


assert all(
    packet[
        "representative_evidence"
    ]
    for packet in (
        review_packets_by_slot_id.values()
    )
)


assert all(
    all(
        is_nonempty_value(
            evidence[
                "evidence_quote_exact"
            ]
        )
        for evidence in (
            packet[
                "representative_evidence"
            ]
        )
    )
    for packet in (
        review_packets_by_slot_id.values()
    )
)


assert all(
    all(
        is_nonempty_value(
            evidence[
                "locator"
            ]
        )
        for evidence in (
            packet[
                "representative_evidence"
            ]
        )
    )
    for packet in (
        review_packets_by_slot_id.values()
    )
)


# =========================================================
# Packet size diagnostics
# =========================================================

review_packet_character_lengths = []


for packet in (
    review_packets_by_slot_id.values()
):

    packet_json = json.dumps(
        packet,
        ensure_ascii=False,
        sort_keys=True,
    )


    review_packet_character_lengths.append(
        len(
            packet_json
        )
    )


review_packet_character_profile = (
    numeric_profile(
        review_packet_character_lengths
    )
)


# =========================================================
# Data-driven Pilot Strata
#
# Exact observed combination:
# Authority
# + representative Source Type signature
# + representative Claim Type signature
# + actual Review Risk signature
# =========================================================

def pilot_stratum_key(
    slot,
):

    representative = (
        slot[
            "representative_claim"
        ]
    )


    authority_signature = (
        representative[
            "authority_class"
        ]
    )


    source_type_signature = (
        " + ".join(
            sorted(
                representative[
                    "source_types"
                ]
            )
        )
    )


    claim_type_signature = (
        " + ".join(
            sorted(
                representative[
                    "claim_types"
                ]
            )
        )
    )


    risk_signature = (
        " + ".join(
            sorted(
                slot[
                    "review_flags"
                ]
            )
        )
        if (
            slot[
                "review_flags"
            ]
        )
        else "NO_REVIEW_FLAGS"
    )


    return (
        authority_signature,
        source_type_signature,
        claim_type_signature,
        risk_signature,
    )


pilot_strata = defaultdict(
    list
)


for slot in (
    provisional_seed_slot_records
):

    stratum_key = (
        pilot_stratum_key(
            slot
        )
    )


    pilot_strata[
        stratum_key
    ].append(
        slot[
            "seed_slot_id"
        ]
    )


# =========================================================
# Deterministic challenge-oriented representative
#
# Within each observed stratum:
# 1. more external relation context
# 2. larger Semantic-Duplicate component
# 3. more representative Evidence
# 4. shorter Claim as ambiguity stressor
# 5. seed_slot_id tie-break
# =========================================================

def pilot_candidate_rank(
    seed_slot_id,
):

    slot = (
        seed_slot_by_id[
            seed_slot_id
        ]
    )


    relation_context_count = len(
        external_relation_context_by_slot[
            seed_slot_id
        ]
    )


    member_claim_count = (
        slot[
            "member_claim_count"
        ]
    )


    representative_evidence_count = (
        slot[
            "representative_claim"
        ][
            "evidence_count"
        ]
    )


    claim_text_length = len(
        slot[
            "representative_claim"
        ][
            "claim_text_fa"
        ]
    )


    return (
        -relation_context_count,
        -member_claim_count,
        -representative_evidence_count,
        claim_text_length,
        seed_slot_id,
    )


pilot_seed_slot_ids = []


pilot_selection_records = []


for stratum_key in sorted(
    pilot_strata
):

    stratum_member_slot_ids = sorted(
        pilot_strata[
            stratum_key
        ]
    )


    selected_seed_slot_id = min(
        stratum_member_slot_ids,
        key=pilot_candidate_rank,
    )


    pilot_seed_slot_ids.append(
        selected_seed_slot_id
    )


    pilot_selection_records.append({
        "pilot_selection_version": (
            PILOT_SELECTION_VERSION
        ),

        "authority_signature": (
            stratum_key[
                0
            ]
        ),

        "source_type_signature": (
            stratum_key[
                1
            ]
        ),

        "claim_type_signature": (
            stratum_key[
                2
            ]
        ),

        "risk_signature": (
            stratum_key[
                3
            ]
        ),

        "stratum_population": (
            len(
                stratum_member_slot_ids
            )
        ),

        "selected_seed_slot_id": (
            selected_seed_slot_id
        ),

        "selected_review_packet_sha256": (
            review_packets_by_slot_id[
                selected_seed_slot_id
            ][
                "review_packet_sha256"
            ]
        ),
    })


# =========================================================
# Pilot validation
# =========================================================

observed_pilot_strata_count = len(
    pilot_strata
)


pilot_size = len(
    pilot_seed_slot_ids
)


assert (
    pilot_size
    == observed_pilot_strata_count
)


assert (
    len(
        pilot_seed_slot_ids
    )
    == len(
        set(
            pilot_seed_slot_ids
        )
    )
)


assert all(
    seed_slot_id
    in review_packets_by_slot_id
    for seed_slot_id in (
        pilot_seed_slot_ids
    )
)


# Every observed stratum must be represented exactly once.
selected_stratum_keys = [
    pilot_stratum_key(
        seed_slot_by_id[
            seed_slot_id
        ]
    )
    for seed_slot_id in (
        pilot_seed_slot_ids
    )
]


assert (
    len(
        selected_stratum_keys
    )
    == len(
        set(
            selected_stratum_keys
        )
    )
)


assert (
    set(
        selected_stratum_keys
    )
    == set(
        pilot_strata
    )
)


# =========================================================
# Pilot coverage diagnostics
# =========================================================

pilot_slots = [
    seed_slot_by_id[
        seed_slot_id
    ]
    for seed_slot_id in (
        pilot_seed_slot_ids
    )
]


pilot_authority_counts = Counter(
    slot[
        "representative_claim"
    ][
        "authority_class"
    ]
    for slot in (
        pilot_slots
    )
)


pilot_source_type_counts = Counter(
    source_type
    for slot in (
        pilot_slots
    )
    for source_type in (
        slot[
            "representative_claim"
        ][
            "source_types"
        ]
    )
)


pilot_claim_type_counts = Counter(
    claim_type
    for slot in (
        pilot_slots
    )
    for claim_type in (
        slot[
            "representative_claim"
        ][
            "claim_types"
        ]
    )
)


pilot_review_flag_counts = Counter(
    flag
    for slot in (
        pilot_slots
    )
    for flag in (
        slot[
            "review_flags"
        ]
    )
)


# All observed categories must appear in Pilot.
assert (
    set(
        pilot_authority_counts
    )
    == set(
        representative_authority_counts
    )
)


assert (
    set(
        pilot_source_type_counts
    )
    == set(
        representative_source_type_counts
    )
)


assert (
    set(
        pilot_claim_type_counts
    )
    == set(
        representative_claim_type_counts
    )
)


assert (
    set(
        pilot_review_flag_counts
    )
    == set(
        slot_review_flag_counts
    )
)


# =========================================================
# Full Packet Manifest Hash
# =========================================================

review_packet_hash_manifest = [
    {
        "seed_slot_id": (
            seed_slot_id
        ),

        "review_packet_sha256": (
            review_packets_by_slot_id[
                seed_slot_id
            ][
                "review_packet_sha256"
            ]
        ),
    }
    for seed_slot_id in sorted(
        review_packets_by_slot_id
    )
]


review_packet_hash_manifest_text = (
    json.dumps(
        review_packet_hash_manifest,
        ensure_ascii=False,
        sort_keys=True,
        separators=(
            ",",
            ":",
        ),
    )
)


review_packet_hash_manifest_sha256 = (
    hashlib.sha256(
        review_packet_hash_manifest_text.encode(
            "utf-8"
        )
    ).hexdigest()
)


# =========================================================
# Pilot Selection Hash
# =========================================================

pilot_selection_text = json.dumps(
    pilot_selection_records,
    ensure_ascii=False,
    sort_keys=True,
    separators=(
        ",",
        ":",
    ),
)


pilot_selection_sha256 = (
    hashlib.sha256(
        pilot_selection_text.encode(
            "utf-8"
        )
    ).hexdigest()
)


# =========================================================
# Stage Gate
# =========================================================

review_packet_and_pilot_design_status = (
    "PASS"
)


pilot_runtime_preflight_readiness = (
    "READY_FOR_SEMANTIC_REVIEW_PILOT_RUNTIME_PREFLIGHT"
)


# =========================================================
# Result
# =========================================================

print(
    "=== NB12j REVIEW PACKET / STRATIFIED PILOT DESIGN ==="
)

print()

print(
    "Review Packets built:",
    len(
        review_packets_by_slot_id
    ),
)

print(
    "Review Packet character profile:",
    review_packet_character_profile,
)

print(
    "Review Packet manifest SHA-256:",
    review_packet_hash_manifest_sha256,
)


print()

print(
    "Observed Pilot Strata:",
    observed_pilot_strata_count,
)

print(
    "Pilot Slots selected:",
    pilot_size,
)

print(
    "Pilot fraction of Full Review:",
    round(
        pilot_size
        / semantic_review_population,
        4,
    ),
)


print()

print(
    "Pilot Authority coverage:",
    dict(
        sorted(
            pilot_authority_counts.items()
        )
    ),
)

print(
    "Pilot Source-Type coverage:",
    dict(
        sorted(
            pilot_source_type_counts.items()
        )
    ),
)

print(
    "Pilot Claim-Type coverage:",
    dict(
        sorted(
            pilot_claim_type_counts.items()
        )
    ),
)

print(
    "Pilot Review-Flag coverage:",
    dict(
        sorted(
            pilot_review_flag_counts.items()
        )
    ),
)


print()

print(
    "All observed Authority classes covered:",
    (
        set(
            pilot_authority_counts
        )
        == set(
            representative_authority_counts
        )
    ),
)

print(
    "All observed Source Types covered:",
    (
        set(
            pilot_source_type_counts
        )
        == set(
            representative_source_type_counts
        )
    ),
)

print(
    "All observed Claim Types covered:",
    (
        set(
            pilot_claim_type_counts
        )
        == set(
            representative_claim_type_counts
        )
    ),
)

print(
    "All observed Review Flags covered:",
    (
        set(
            pilot_review_flag_counts
        )
        == set(
            slot_review_flag_counts
        )
    ),
)


print()

print(
    "Pilot Selection SHA-256:",
    pilot_selection_sha256,
)

print(
    "Pilot fixed target imposed:",
    False,
)

print(
    "API calls performed:",
    0,
)

print(
    "Files modified by this Cell:",
    0,
)


print()

print(
    "Next stage:",
    "SEMANTIC_REVIEW_PILOT_RUNTIME_PREFLIGHT",
)

print(
    "Readiness:",
    pilot_runtime_preflight_readiness,
)


print()

print(
    "CODE CELL 13 GATE:",
    review_packet_and_pilot_design_status,
)
````

````python
# Code Cell 15
# FINAL FULL-3817 SEMANTIC REVIEW
# RERUN-SAFE CONTROLLER REBUILD + CHECKPOINT VALIDATION
#
# IMPORTANT:
# - ZERO API CALLS
# - ZERO PAID EXECUTION
# - Rebuilds exact historical Review controller
# - Preserves original prompt identity / hashes
# - Reuses existing Full + Legacy checkpoints
# - Reconstructs downstream state after Kernel Restart
# =========================================================


from pathlib import Path
from collections import Counter
import hashlib
import json


# =========================================================
# EXECUTION SAFETY
# =========================================================

EXECUTE_FULL_REVIEW = False

assert EXECUTE_FULL_REVIEW is False


# =========================================================
# REQUIRED STABLE UPSTREAM STATE
# =========================================================

required_upstream_names = [
    "project_root",
    "provisional_seed_slot_records",
    "review_packets_by_slot_id",
    "semantic_review_model",
]


missing_upstream_names = [
    name
    for name in required_upstream_names
    if name not in globals()
]


assert not missing_upstream_names, (
    "Missing stable upstream objects: "
    + ", ".join(
        missing_upstream_names
    )
)


# =========================================================
# FROZEN FULL REVIEW CONTRACT
#
# EXACT historical values.
# Do not change: checkpoint identity depends on these.
# =========================================================

FULL_REVIEW_POLICY_VERSION = (
    "golden_v2_full_semantic_review_controller_v1"
)


PROMPT_VERSION = (
    "golden_v2_budget_canary_semantic_judge_v1"
)


MAX_TOKENS = 256

TEMPERATURE = 0.7

TOP_P = 0.8


MAX_ACCEPTABLE_RESIDUAL_REASONING_TOKENS = 1


ALLOWED_VERDICTS = {
    "ACCEPT_SEED",
    "HOLD_REPAIR",
    "HOLD_RELATION_REVIEW",
    "REJECT_AS_GOLDEN_SEED",
}


REQUIRED_DIMENSIONS = {
    "standalone_question_possible",
    "answerable_from_attached_evidence",
    "question_scope_unambiguous",
    "independent_evaluation_target",
    "attribution_preservable",
    "requires_external_unprovided_context",
}


CANONICAL_REASON_CODES = {
    "PASS_STANDALONE",
    "PASS_EVIDENCE_GROUNDED",
    "PASS_UNAMBIGUOUS",
    "PASS_INDEPENDENT",
    "ATTRIBUTION_REQUIRED",
    "INSUFFICIENT_STANDALONE_CONTEXT",
    "QUESTION_SCOPE_AMBIGUOUS",
    "EVIDENCE_SCOPE_MISMATCH",
    "SUBSUMPTION_DEPENDENCY",
    "PARTIAL_OVERLAP_DEPENDENCY",
    "TOO_CONTEXT_DEPENDENT",
    "NOT_BENCHMARK_WORTHY_AS_INDEPENDENT_SEED",
}


# =========================================================
# HISTORICAL PRICING CONSTANTS
#
# Retained for downstream compatibility / reporting.
# NO spending occurs in this Cell.
# =========================================================

ARVAN_REFERENCE_INPUT_TOKENS = 6_656_488

ARVAN_REFERENCE_INPUT_COST_TOMAN = 407_759


ARVAN_REFERENCE_OUTPUT_TOKENS = 4_856_181

ARVAN_REFERENCE_OUTPUT_COST_TOMAN = 892_333


INPUT_TOMAN_PER_TOKEN = (
    ARVAN_REFERENCE_INPUT_COST_TOMAN
    / ARVAN_REFERENCE_INPUT_TOKENS
)


OUTPUT_TOMAN_PER_TOKEN = (
    ARVAN_REFERENCE_OUTPUT_COST_TOMAN
    / ARVAN_REFERENCE_OUTPUT_TOKENS
)


INPUT_TOMAN_PER_MILLION = (
    INPUT_TOMAN_PER_TOKEN
    * 1_000_000
)


OUTPUT_TOMAN_PER_MILLION = (
    OUTPUT_TOMAN_PER_TOKEN
    * 1_000_000
)


FULL_REVIEW_HARD_COST_CAP_TOMAN = 600_000


# =========================================================
# EXACT HISTORICAL SYSTEM PROMPT
#
# Must remain byte-compatible with the successful Review.
# =========================================================

SYSTEM_PROMPT = """
You are a strict Persian Golden Dataset seed judge.

Judge ONLY from the supplied packet.
Do not use external knowledge.
Do not invent facts.
Do not expose reasoning.
Return JSON only.

A valid seed must support at least one independent,
unambiguous benchmark question whose answer is grounded
in the attached evidence.

Allowed verdicts:
ACCEPT_SEED
HOLD_REPAIR
HOLD_RELATION_REVIEW
REJECT_AS_GOLDEN_SEED

Return exactly these four top-level fields:

{
  "seed_slot_id": "...",
  "verdict": "...",
  "dimensions": {
    "standalone_question_possible": true,
    "answerable_from_attached_evidence": true,
    "question_scope_unambiguous": true,
    "independent_evaluation_target": true,
    "attribution_preservable": true,
    "requires_external_unprovided_context": false
  },
  "brief_reason_fa": "..."
}

Do NOT return reason_codes.
Keep brief_reason_fa under 25 Persian words.
""".strip()


# =========================================================
# EXACT HISTORICAL PAYLOAD
# =========================================================

def build_full_review_payload(
    packet,
):

    return {
        "seed_slot_id": (
            packet[
                "seed_slot_id"
            ]
        ),

        "claim": (
            packet[
                "representative_claim"
            ]
        ),

        "evidence": (
            packet[
                "representative_evidence"
            ]
        ),

        "component_context": (
            packet[
                "component_context"
            ]
        ),

        "relation_context": (
            packet[
                "relation_context"
            ]
        ),

        "review_flags": (
            packet[
                "review_flags"
            ]
        ),
    }


def build_full_review_user_prompt(
    packet,
):

    return (
        json.dumps(
            build_full_review_payload(
                packet
            ),
            ensure_ascii=False,
            sort_keys=True,
            separators=(",", ":"),
        )
        + "\n/no_think"
    )


# =========================================================
# REBUILD STABLE FULL POPULATION
# =========================================================

assert (
    len(
        provisional_seed_slot_records
    )
    == 3817
), (
    "Expected exactly 3817 provisional Seed Slots."
)


full_seed_slot_ids = sorted(
    slot[
        "seed_slot_id"
    ]
    for slot in (
        provisional_seed_slot_records
    )
)


assert (
    len(
        full_seed_slot_ids
    )
    == 3817
)


assert (
    len(
        set(
            full_seed_slot_ids
        )
    )
    == 3817
)


assert all(
    seed_slot_id
    in review_packets_by_slot_id
    for seed_slot_id in (
        full_seed_slot_ids
    )
), (
    "At least one Seed is missing its Review Packet."
)


# =========================================================
# EXACT HISTORICAL REQUEST IDENTITY
# =========================================================

def build_request_identity(
    seed_slot_id,
):

    packet = (
        review_packets_by_slot_id[
            seed_slot_id
        ]
    )


    user_prompt = (
        build_full_review_user_prompt(
            packet
        )
    )


    identity_payload = {
        "prompt_version": (
            PROMPT_VERSION
        ),

        "model": (
            semantic_review_model
        ),

        "seed_slot_id": (
            seed_slot_id
        ),

        "system_prompt": (
            SYSTEM_PROMPT
        ),

        "user_prompt": (
            user_prompt
        ),

        "max_tokens": (
            MAX_TOKENS
        ),

        "temperature": (
            TEMPERATURE
        ),

        "top_p": (
            TOP_P
        ),

        "enable_thinking": (
            False
        ),
    }


    identity_text = (
        json.dumps(
            identity_payload,
            ensure_ascii=False,
            sort_keys=True,
            separators=(",", ":"),
        )
    )


    prompt_sha256 = (
        hashlib.sha256(
            identity_text.encode(
                "utf-8"
            )
        ).hexdigest()
    )


    prompt_character_count = (
        len(
            SYSTEM_PROMPT
        )
        + len(
            user_prompt
        )
    )


    return {
        "packet": (
            packet
        ),

        "user_prompt": (
            user_prompt
        ),

        "prompt_sha256": (
            prompt_sha256
        ),

        "prompt_character_count": (
            prompt_character_count
        ),
    }


# =========================================================
# EXACT HISTORICAL CACHE ARCHITECTURE
# =========================================================

legacy_cache_dir = (
    project_root
    / "artifacts"
    / "phase2"
    / "golden_dataset_v2"
    / "work"
    / "budget_canary_v1"
)


full_cache_dir = (
    project_root
    / "artifacts"
    / "phase2"
    / "golden_dataset_v2"
    / "work"
    / "semantic_review_full_v1"
)


def checkpoint_filename(
    seed_slot_id,
    prompt_sha256,
):

    return (
        f"{seed_slot_id}"
        f"__{prompt_sha256[:16]}"
        ".json"
    )


def exact_checkpoint_candidates(
    seed_slot_id,
    prompt_sha256,
):

    filename = (
        checkpoint_filename(
            seed_slot_id,
            prompt_sha256,
        )
    )


    return [
        full_cache_dir
        / filename,

        legacy_cache_dir
        / filename,
    ]


def locate_exact_checkpoint(
    seed_slot_id,
    prompt_sha256,
):

    for path in (
        exact_checkpoint_candidates(
            seed_slot_id,
            prompt_sha256,
        )
    ):

        if path.exists():

            return path


    return None


# =========================================================
# EXACT HISTORICAL VALIDATION CONTRACT
# =========================================================

def validate_cached_or_runtime_record(
    seed_slot_id,
    record,
):

    parsed_json = (
        record.get(
            "parsed_json"
        )
    )


    usage = (
        record.get(
            "usage"
        )
        or {}
    )


    reasoning_tokens = int(
        usage.get(
            "reasoning_tokens",
            0,
        )
        or 0
    )


    finish_reason = (
        record.get(
            "finish_reason"
        )
    )


    checks = {
        "json_object": (
            isinstance(
                parsed_json,
                dict,
            )
        ),

        "top_level_fields_exact": (
            isinstance(
                parsed_json,
                dict,
            )
            and set(
                parsed_json.keys()
            )
            == {
                "seed_slot_id",
                "verdict",
                "dimensions",
                "brief_reason_fa",
            }
        ),

        "seed_slot_matches": (
            isinstance(
                parsed_json,
                dict,
            )
            and parsed_json.get(
                "seed_slot_id"
            )
            == seed_slot_id
        ),

        "verdict_allowed": (
            isinstance(
                parsed_json,
                dict,
            )
            and parsed_json.get(
                "verdict"
            )
            in ALLOWED_VERDICTS
        ),

        "dimensions_exact": (
            isinstance(
                parsed_json,
                dict,
            )
            and isinstance(
                parsed_json.get(
                    "dimensions"
                ),
                dict,
            )
            and set(
                parsed_json[
                    "dimensions"
                ].keys()
            )
            == REQUIRED_DIMENSIONS
            and all(
                isinstance(
                    value,
                    bool,
                )
                for value in (
                    parsed_json[
                        "dimensions"
                    ].values()
                )
            )
        ),

        "brief_reason_present": (
            isinstance(
                parsed_json,
                dict,
            )
            and isinstance(
                parsed_json.get(
                    "brief_reason_fa"
                ),
                str,
            )
            and bool(
                parsed_json.get(
                    "brief_reason_fa",
                    "",
                ).strip()
            )
        ),

        "not_truncated": (
            finish_reason
            != "length"
        ),

        "residual_reasoning_within_tolerance": (
            0
            <= reasoning_tokens
            <= MAX_ACCEPTABLE_RESIDUAL_REASONING_TOKENS
        ),
    }


    accept_consistency = True


    if (
        isinstance(
            parsed_json,
            dict,
        )
        and parsed_json.get(
            "verdict"
        )
        == "ACCEPT_SEED"
        and isinstance(
            parsed_json.get(
                "dimensions"
            ),
            dict,
        )
    ):

        dimensions = (
            parsed_json[
                "dimensions"
            ]
        )


        accept_consistency = (
            dimensions.get(
                "standalone_question_possible"
            )
            is True
            and dimensions.get(
                "answerable_from_attached_evidence"
            )
            is True
            and dimensions.get(
                "question_scope_unambiguous"
            )
            is True
            and dimensions.get(
                "independent_evaluation_target"
            )
            is True
            and dimensions.get(
                "attribution_preservable"
            )
            is True
            and dimensions.get(
                "requires_external_unprovided_context"
            )
            is False
        )


    checks[
        "accept_verdict_dimension_consistency"
    ] = (
        accept_consistency
    )


    return (
        checks,
        (
            "PASS"
            if all(
                checks.values()
            )
            else "FAIL"
        ),
    )


# =========================================================
# EXACT DETERMINISTIC LOCAL REASON CODES
# =========================================================

def derive_canonical_reason_codes(
    dimensions,
    review_flags,
):

    codes = []


    if dimensions[
        "standalone_question_possible"
    ]:

        codes.append(
            "PASS_STANDALONE"
        )

    else:

        codes.append(
            "INSUFFICIENT_STANDALONE_CONTEXT"
        )


    if dimensions[
        "answerable_from_attached_evidence"
    ]:

        codes.append(
            "PASS_EVIDENCE_GROUNDED"
        )

    else:

        codes.append(
            "EVIDENCE_SCOPE_MISMATCH"
        )


    if dimensions[
        "question_scope_unambiguous"
    ]:

        codes.append(
            "PASS_UNAMBIGUOUS"
        )

    else:

        codes.append(
            "QUESTION_SCOPE_AMBIGUOUS"
        )


    if dimensions[
        "independent_evaluation_target"
    ]:

        codes.append(
            "PASS_INDEPENDENT"
        )

    else:

        codes.append(
            "NOT_BENCHMARK_WORTHY_AS_INDEPENDENT_SEED"
        )


    if (
        "ATTRIBUTION_REQUIRED"
        in review_flags
    ):

        codes.append(
            "ATTRIBUTION_REQUIRED"
        )


    if dimensions[
        "requires_external_unprovided_context"
    ]:

        codes.append(
            "TOO_CONTEXT_DEPENDENT"
        )


    if not dimensions[
        "independent_evaluation_target"
    ]:

        if (
            "SUBSUMPTION_REVIEW"
            in review_flags
        ):

            codes.append(
                "SUBSUMPTION_DEPENDENCY"
            )


        if (
            "PARTIAL_OVERLAP_REVIEW"
            in review_flags
        ):

            codes.append(
                "PARTIAL_OVERLAP_DEPENDENCY"
            )


    codes = list(
        dict.fromkeys(
            codes
        )
    )


    assert all(
        code
        in CANONICAL_REASON_CODES
        for code in codes
    )


    return codes


# =========================================================
# SAFE JSON READ
# =========================================================

def safe_read_json(
    path,
):

    try:

        return json.loads(
            path.read_text(
                encoding="utf-8"
            )
        )

    except Exception:

        return None


# =========================================================
# BUILD EXACT REQUEST MANIFEST
# =========================================================

request_manifest = []


for seed_slot_id in (
    full_seed_slot_ids
):

    identity = (
        build_request_identity(
            seed_slot_id
        )
    )


    checkpoint_path = (
        locate_exact_checkpoint(
            seed_slot_id,
            identity[
                "prompt_sha256"
            ],
        )
    )


    request_manifest.append({
        "seed_slot_id": (
            seed_slot_id
        ),

        "prompt_sha256": (
            identity[
                "prompt_sha256"
            ]
        ),

        "prompt_character_count": (
            identity[
                "prompt_character_count"
            ]
        ),

        "checkpoint_path": (
            checkpoint_path
        ),
    })


assert (
    len(
        request_manifest
    )
    == 3817
)


manifest_by_seed_id = {
    item[
        "seed_slot_id"
    ]: item
    for item in (
        request_manifest
    )
}


# =========================================================
# READ + VALIDATE ALL 3817 CHECKPOINTS
# =========================================================

final_valid_records = {}

final_invalid_records = []

final_unreadable_records = []

final_missing_seed_slot_ids = []


for item in (
    request_manifest
):

    seed_slot_id = (
        item[
            "seed_slot_id"
        ]
    )


    checkpoint_path = (
        item[
            "checkpoint_path"
        ]
    )


    # -----------------------------------------------------
    # Missing exact checkpoint
    # -----------------------------------------------------

    if checkpoint_path is None:

        final_missing_seed_slot_ids.append(
            seed_slot_id
        )

        continue


    # -----------------------------------------------------
    # Read exact checkpoint
    # -----------------------------------------------------

    record = (
        safe_read_json(
            checkpoint_path
        )
    )


    if not isinstance(
        record,
        dict,
    ):

        final_unreadable_records.append({
            "seed_slot_id": (
                seed_slot_id
            ),

            "checkpoint_path": (
                str(
                    checkpoint_path
                )
            ),
        })

        continue


    # -----------------------------------------------------
    # Exact identity validation
    # -----------------------------------------------------

    if (
        record.get(
            "prompt_sha256"
        )
        != item[
            "prompt_sha256"
        ]
    ):

        final_invalid_records.append({
            "seed_slot_id": (
                seed_slot_id
            ),

            "checkpoint_path": (
                str(
                    checkpoint_path
                )
            ),

            "reason": (
                "PROMPT_SHA256_MISMATCH"
            ),
        })

        continue


    if (
        record.get(
            "seed_slot_id"
        )
        != seed_slot_id
    ):

        final_invalid_records.append({
            "seed_slot_id": (
                seed_slot_id
            ),

            "checkpoint_path": (
                str(
                    checkpoint_path
                )
            ),

            "reason": (
                "SEED_SLOT_ID_MISMATCH"
            ),
        })

        continue


    # -----------------------------------------------------
    # Semantic Review contract validation
    # -----------------------------------------------------

    try:

        (
            validation_checks,
            validation_status,
        ) = (
            validate_cached_or_runtime_record(
                seed_slot_id,
                record,
            )
        )

    except Exception as exc:

        final_invalid_records.append({
            "seed_slot_id": (
                seed_slot_id
            ),

            "checkpoint_path": (
                str(
                    checkpoint_path
                )
            ),

            "reason": (
                "VALIDATOR_EXCEPTION"
            ),

            "error": (
                f"{type(exc).__name__}: {exc}"
            ),
        })

        continue


    if (
        validation_status
        != "PASS"
    ):

        failed_checks = [
            name
            for (
                name,
                value,
            ) in (
                validation_checks.items()
            )
            if not value
        ]


        final_invalid_records.append({
            "seed_slot_id": (
                seed_slot_id
            ),

            "checkpoint_path": (
                str(
                    checkpoint_path
                )
            ),

            "reason": (
                "CHECKPOINT_VALIDATION_FAIL"
            ),

            "failed_checks": (
                failed_checks
            ),
        })

        continue


    # -----------------------------------------------------
    # Rebuild deterministic local reason codes in memory.
    #
    # IMPORTANT:
    # Checkpoint file is NOT modified.
    # -----------------------------------------------------

    normalized_record = dict(
        record
    )


    normalized_record[
        "current_validation_checks"
    ] = (
        validation_checks
    )


    normalized_record[
        "current_validation_status"
    ] = (
        validation_status
    )


    parsed_json = (
        normalized_record[
            "parsed_json"
        ]
    )


    normalized_record[
        "canonical_reason_codes"
    ] = (
        derive_canonical_reason_codes(
            dimensions=(
                parsed_json[
                    "dimensions"
                ]
            ),

            review_flags=(
                review_packets_by_slot_id[
                    seed_slot_id
                ][
                    "review_flags"
                ]
            ),
        )
    )


    normalized_record[
        "reason_code_source"
    ] = (
        "DETERMINISTIC_LOCAL_DERIVATION"
    )


    final_valid_records[
        seed_slot_id
    ] = (
        normalized_record
    )


# =========================================================
# POPULATION ACCOUNTING
# =========================================================

final_accounted_population = (
    len(
        final_valid_records
    )
    + len(
        final_missing_seed_slot_ids
    )
    + len(
        final_invalid_records
    )
    + len(
        final_unreadable_records
    )
)


assert (
    final_accounted_population
    == 3817
), (
    "Final checkpoint accounting must equal 3817."
)


# =========================================================
# DOWNSTREAM-COMPATIBLE CACHE STATE
# =========================================================

valid_cached_records = [
    {
        "seed_slot_id": (
            seed_slot_id
        ),

        "checkpoint_path": (
            manifest_by_seed_id[
                seed_slot_id
            ][
                "checkpoint_path"
            ]
        ),

        "record": (
            record
        ),

        "validation_checks": (
            record[
                "current_validation_checks"
            ]
        ),

        "validation_status": (
            "PASS"
        ),

        "prompt_character_count": (
            manifest_by_seed_id[
                seed_slot_id
            ][
                "prompt_character_count"
            ]
        ),
    }
    for (
        seed_slot_id,
        record,
    ) in sorted(
        final_valid_records.items()
    )
]


invalid_cached_records = list(
    final_invalid_records
)


cached_seed_slot_ids = set(
    final_valid_records
)


missing_seed_slot_ids = sorted(
    set(
        full_seed_slot_ids
    )
    - cached_seed_slot_ids
)


# =========================================================
# DOWNSTREAM POST-REVIEW STATE
# =========================================================

post_review_records_by_seed_id = dict(
    final_valid_records
)


post_review_missing_seed_ids = list(
    final_missing_seed_slot_ids
)


post_review_invalid_records = list(
    final_invalid_records
)


post_review_unreadable_records = list(
    final_unreadable_records
)


# =========================================================
# AGGREGATES
# =========================================================

final_verdict_counts = Counter(
    record[
        "parsed_json"
    ][
        "verdict"
    ]
    for record in (
        final_valid_records.values()
    )
)


final_reasoning_distribution = Counter(
    int(
        (
            record.get(
                "usage"
            )
            or {}
        ).get(
            "reasoning_tokens",
            0,
        )
        or 0
    )
    for record in (
        final_valid_records.values()
    )
)


final_429_retry_distribution = Counter(
    int(
        record.get(
            "429_retry_count",
            0,
        )
        or 0
    )
    for record in (
        final_valid_records.values()
    )
)


final_input_tokens = sum(
    int(
        (
            record.get(
                "usage"
            )
            or {}
        ).get(
            "input_tokens",
            0,
        )
        or 0
    )
    for record in (
        final_valid_records.values()
    )
)


final_output_tokens = sum(
    int(
        (
            record.get(
                "usage"
            )
            or {}
        ).get(
            "output_tokens",
            0,
        )
        or 0
    )
    for record in (
        final_valid_records.values()
    )
)


final_total_tokens = sum(
    int(
        (
            record.get(
                "usage"
            )
            or {}
        ).get(
            "total_tokens",
            0,
        )
        or 0
    )
    for record in (
        final_valid_records.values()
    )
)


def estimated_cost_toman(
    input_tokens,
    output_tokens,
):

    return (
        input_tokens
        * INPUT_TOMAN_PER_TOKEN
        + output_tokens
        * OUTPUT_TOMAN_PER_TOKEN
    )


final_estimated_total_cost_toman = (
    estimated_cost_toman(
        final_input_tokens,
        final_output_tokens,
    )
)


# =========================================================
# VERDICT-DERIVED POPULATIONS
# =========================================================

accepted_seed_slot_ids = sorted(
    seed_slot_id
    for (
        seed_slot_id,
        record,
    ) in (
        final_valid_records.items()
    )
    if (
        record[
            "parsed_json"
        ][
            "verdict"
        ]
        == "ACCEPT_SEED"
    )
)


hold_repair_seed_slot_ids = sorted(
    seed_slot_id
    for (
        seed_slot_id,
        record,
    ) in (
        final_valid_records.items()
    )
    if (
        record[
            "parsed_json"
        ][
            "verdict"
        ]
        == "HOLD_REPAIR"
    )
)


hold_relation_review_seed_slot_ids = sorted(
    seed_slot_id
    for (
        seed_slot_id,
        record,
    ) in (
        final_valid_records.items()
    )
    if (
        record[
            "parsed_json"
        ][
            "verdict"
        ]
        == "HOLD_RELATION_REVIEW"
    )
)


rejected_seed_slot_ids = sorted(
    seed_slot_id
    for (
        seed_slot_id,
        record,
    ) in (
        final_valid_records.items()
    )
    if (
        record[
            "parsed_json"
        ][
            "verdict"
        ]
        == "REJECT_AS_GOLDEN_SEED"
    )
)


# =========================================================
# FINAL RERUN-SAFE GATE
# =========================================================

full_review_complete = (
    len(
        final_valid_records
    )
    == 3817
    and len(
        final_missing_seed_slot_ids
    )
    == 0
    and len(
        final_invalid_records
    )
    == 0
    and len(
        final_unreadable_records
    )
    == 0
)


full_review_execution_status = (
    "PASS"
    if full_review_complete
    else "INCOMPLETE_EXISTING_CHECKPOINT_SET"
)


full_review_next_readiness = (
    "READY_FOR_GOLDEN_V2_POST_REVIEW_ADJUDICATION"
    if full_review_complete
    else "BLOCKED_REVIEW_CHECKPOINT_REPAIR_REQUIRED"
)


# =========================================================
# STRICT SAFETY:
#
# Missing/invalid Review must NEVER silently trigger paid
# execution during Restart / Run All.
# =========================================================

api_calls_performed = 0

files_modified = 0


# =========================================================
# OUTPUT
# =========================================================

print(
    "=== NB12j RERUN-SAFE FULL-3817 REVIEW VALIDATION ==="
)

print()

print(
    "Execution mode:",
    "CHECKPOINT_VALIDATION_ONLY",
)

print(
    "API calls performed:",
    api_calls_performed,
)

print(
    "Files modified:",
    files_modified,
)


print()

print(
    "Target population:",
    len(
        full_seed_slot_ids
    ),
)

print(
    "Valid exact reviews:",
    len(
        final_valid_records
    ),
)

print(
    "Missing exact reviews:",
    len(
        final_missing_seed_slot_ids
    ),
)

print(
    "Invalid reviews:",
    len(
        final_invalid_records
    ),
)

print(
    "Unreadable reviews:",
    len(
        final_unreadable_records
    ),
)


print()

print(
    "Verdict counts:",
    dict(
        sorted(
            final_verdict_counts.items()
        )
    ),
)

print(
    "Residual reasoning distribution:",
    dict(
        sorted(
            final_reasoning_distribution.items()
        )
    ),
)

print(
    "429 retry distribution:",
    dict(
        sorted(
            final_429_retry_distribution.items()
        )
    ),
)


print()

print(
    "VALID checkpoint input tokens:",
    final_input_tokens,
)

print(
    "VALID checkpoint output tokens:",
    final_output_tokens,
)

print(
    "VALID checkpoint total tokens:",
    final_total_tokens,
)

print(
    "Estimated historical review cost:",
    round(
        final_estimated_total_cost_toman
    ),
    "toman",
)


print()

print(
    "Post-review populations:"
)

print(
    " - Accepted:",
    len(
        accepted_seed_slot_ids
    ),
)

print(
    " - Hold Repair:",
    len(
        hold_repair_seed_slot_ids
    ),
)

print(
    " - Hold Relation Review:",
    len(
        hold_relation_review_seed_slot_ids
    ),
)

print(
    " - Rejected:",
    len(
        rejected_seed_slot_ids
    ),
)


if final_missing_seed_slot_ids:

    print()

    print(
        "First missing Seed IDs:",
        final_missing_seed_slot_ids[
            :10
        ],
    )


if final_invalid_records:

    print()

    print(
        "First invalid Review records:"
    )

    for item in (
        final_invalid_records[
            :10
        ]
    ):

        print(
            " -",
            item,
        )


if final_unreadable_records:

    print()

    print(
        "First unreadable Review records:"
    )

    for item in (
        final_unreadable_records[
            :10
        ]
    ):

        print(
            " -",
            item,
        )


print()

print(
    "Frozen Golden artifacts modified:",
    False,
)

print(
    "Paid Full Review executed:",
    False,
)

print(
    "Next readiness:",
    full_review_next_readiness,
)

print(
    "CODE CELL 15 GATE:",
    full_review_execution_status,
)


# =========================================================
# HARD STOP ONLY AFTER DIAGNOSTIC OUTPUT
# =========================================================

assert full_review_complete, (
    "The historical Full Review checkpoint set is not "
    "complete/valid. No API calls were made. "
    "Repair the checkpoint set before continuing."
)
````

<h2>Post-Review Adjudication — بررسی اولیه نتایج Semantic Review</h2>

<p>
اجرای کامل Semantic Review برای جمعیت 3817 Seed با موفقیت پایان یافته است.
تمام 3817 Seed دارای Review معتبر هستند و Full Review Gate با وضعیت
<code>PASS</code> بسته شده است.

<p>
در این مرحله وارد بخش <code>Post-Review Adjudication</code> می‌شویم.
هدف این Step، بازیابی و بررسی نتایج واقعی Semantic Review از روی
Checkpointهای ذخیره‌شده و آماده‌سازی جمعیت نهایی برای مرحله Adjudication است.
</p>

<p>
در این بررسی، تعداد Reviewهای معتبر، Missing یا Invalid بودن Reviewها،
توزیع Verdictها و جمعیت Seedهای
<code>ACCEPT_SEED</code>،
<code>HOLD_REPAIR</code>،
<code>HOLD_RELATION_REVIEW</code>

به‌صورت مستقل محاسبه و Validate می‌شود.

این Step فقط عملیات <code>Discovery</code> و <code>Validation</code> انجام می‌دهد.
هیچ API Call جدیدی انجام نمی‌شود، هیچ Checkpoint موجودی تغییر نمی‌کند
و هنوز هیچ Golden Dataset v2 نهایی نوشته نمی‌شود.

<p>
این مرحله زمانی موفق است که دقیقاً 3817 Review معتبر بازیابی شود،
هیچ Review مفقود، خراب یا غیرقابل‌خواندن وجود نداشته باشد،
تمام Verdictها مطابق قرارداد باشند و مجموع جمعیت Verdictها دقیقاً
برابر با 3817 باشد.
</p>

<p>
در صورت PASS شدن این مرحله، جمعیت واقعی Post-Review تثبیت می‌شود

````python
# code cell 16

# =========================================================
# NB12j — POST-REVIEW ADJUDICATION
# STEP 01 — FULL REVIEW DISCOVERY & PREFLIGHT
#
# READ-ONLY
# NO API CALL
# NO ARTIFACT WRITE
# =========================================================

from collections import Counter, defaultdict
import json


# =========================================================
# REQUIRED EXISTING STATE
# =========================================================

required_names = [
    "full_seed_slot_ids",
    "build_request_identity",
    "locate_exact_checkpoint",
    "validate_cached_or_runtime_record",
]


missing_names = [
    name
    for name in required_names
    if name not in globals()
]


assert not missing_names, (
    "Missing required NB12j state: "
    + ", ".join(missing_names)
)


assert len(full_seed_slot_ids) == 3817
assert len(set(full_seed_slot_ids)) == 3817


# =========================================================
# DISCOVER VALID REVIEW RECORDS
# =========================================================

post_review_records_by_seed_id = {}

post_review_missing_seed_ids = []

post_review_invalid_records = []

post_review_unreadable_records = []


for seed_slot_id in full_seed_slot_ids:

    identity = build_request_identity(
        seed_slot_id
    )

    prompt_sha256 = identity[
        "prompt_sha256"
    ]

    checkpoint_path = locate_exact_checkpoint(
        seed_slot_id,
        prompt_sha256,
    )


    # -----------------------------------------------------
    # Missing checkpoint
    # -----------------------------------------------------

    if checkpoint_path is None:

        post_review_missing_seed_ids.append(
            seed_slot_id
        )

        continue


    # -----------------------------------------------------
    # Read checkpoint
    # -----------------------------------------------------

    try:

        record = json.loads(
            checkpoint_path.read_text(
                encoding="utf-8"
            )
        )

    except Exception as exc:

        post_review_unreadable_records.append({
            "seed_slot_id": seed_slot_id,
            "checkpoint_path": str(
                checkpoint_path
            ),
            "error": (
                f"{type(exc).__name__}: {exc}"
            ),
        })

        continue


    # -----------------------------------------------------
    # Validate checkpoint
    # -----------------------------------------------------

    try:

        checks, status = (
            validate_cached_or_runtime_record(
                seed_slot_id,
                record,
            )
        )

    except Exception as exc:

        post_review_invalid_records.append({
            "seed_slot_id": seed_slot_id,
            "checkpoint_path": str(
                checkpoint_path
            ),
            "reason": (
                "VALIDATOR_EXCEPTION"
            ),
            "error": (
                f"{type(exc).__name__}: {exc}"
            ),
        })

        continue


    if status != "PASS":

        failed_checks = [
            name
            for name, value
            in checks.items()
            if not value
        ]

        post_review_invalid_records.append({
            "seed_slot_id": seed_slot_id,
            "checkpoint_path": str(
                checkpoint_path
            ),
            "reason": (
                "VALIDATION_FAIL"
            ),
            "failed_checks": (
                failed_checks
            ),
        })

        continue


    # -----------------------------------------------------
    # Valid record
    # -----------------------------------------------------

    post_review_records_by_seed_id[
        seed_slot_id
    ] = record


# =========================================================
# POPULATION VALIDATION
# =========================================================

valid_review_count = len(
    post_review_records_by_seed_id
)


assert (
    valid_review_count
    + len(post_review_missing_seed_ids)
    + len(post_review_invalid_records)
    + len(post_review_unreadable_records)
    == 3817
), (
    "Review population accounting does not equal 3817."
)


# =========================================================
# VERDICT EXTRACTION
# =========================================================

verdict_to_seed_ids = defaultdict(
    list
)


verdict_extraction_errors = []


for (
    seed_slot_id,
    record,
) in post_review_records_by_seed_id.items():

    parsed_json = record.get(
        "parsed_json"
    )


    if not isinstance(
        parsed_json,
        dict,
    ):

        verdict_extraction_errors.append({
            "seed_slot_id": seed_slot_id,
            "reason": (
                "PARSED_JSON_NOT_OBJECT"
            ),
        })

        continue


    verdict = parsed_json.get(
        "verdict"
    )


    if not isinstance(
        verdict,
        str,
    ):

        verdict_extraction_errors.append({
            "seed_slot_id": seed_slot_id,
            "reason": (
                "VERDICT_MISSING_OR_INVALID"
            ),
        })

        continue


    verdict_to_seed_ids[
        verdict
    ].append(
        seed_slot_id
    )


# =========================================================
# EXPECTED VERDICT CONTRACT
# =========================================================

allowed_verdicts = {
    "ACCEPT_SEED",
    "HOLD_REPAIR",
    "HOLD_RELATION_REVIEW",
    "REJECT_AS_GOLDEN_SEED",
}


observed_verdicts = set(
    verdict_to_seed_ids.keys()
)


unexpected_verdicts = (
    observed_verdicts
    - allowed_verdicts
)


verdict_counts = Counter({
    verdict: len(seed_ids)
    for (
        verdict,
        seed_ids,
    ) in verdict_to_seed_ids.items()
})


total_extracted_verdicts = sum(
    verdict_counts.values()
)


# =========================================================
# DERIVED REVIEW POPULATIONS
# =========================================================

accepted_seed_slot_ids = sorted(
    verdict_to_seed_ids.get(
        "ACCEPT_SEED",
        [],
    )
)


hold_repair_seed_slot_ids = sorted(
    verdict_to_seed_ids.get(
        "HOLD_REPAIR",
        [],
    )
)


hold_relation_review_seed_slot_ids = sorted(
    verdict_to_seed_ids.get(
        "HOLD_RELATION_REVIEW",
        [],
    )
)


rejected_seed_slot_ids = sorted(
    verdict_to_seed_ids.get(
        "REJECT_AS_GOLDEN_SEED",
        [],
    )
)


# =========================================================
# PREFLIGHT GATE
# =========================================================

post_review_preflight_checks = {
    "target_population_is_3817": (
        len(full_seed_slot_ids)
        == 3817
    ),

    "all_3817_reviews_valid": (
        valid_review_count
        == 3817
    ),

    "no_missing_reviews": (
        len(
            post_review_missing_seed_ids
        )
        == 0
    ),

    "no_invalid_reviews": (
        len(
            post_review_invalid_records
        )
        == 0
    ),

    "no_unreadable_reviews": (
        len(
            post_review_unreadable_records
        )
        == 0
    ),

    "no_verdict_extraction_errors": (
        len(
            verdict_extraction_errors
        )
        == 0
    ),

    "no_unexpected_verdicts": (
        len(
            unexpected_verdicts
        )
        == 0
    ),

    "verdict_population_is_3817": (
        total_extracted_verdicts
        == 3817
    ),
}


post_review_preflight_pass = all(
    post_review_preflight_checks.values()
)


# =========================================================
# OUTPUT
# =========================================================

print(
    "=== NB12j POST-REVIEW ADJUDICATION — STEP 01 ==="
)

print()

print(
    "Target Seed population:",
    len(full_seed_slot_ids),
)

print(
    "Valid Semantic Reviews:",
    valid_review_count,
)

print(
    "Missing Reviews:",
    len(
        post_review_missing_seed_ids
    ),
)

print(
    "Invalid Reviews:",
    len(
        post_review_invalid_records
    ),
)

print(
    "Unreadable Reviews:",
    len(
        post_review_unreadable_records
    ),
)

print(
    "Verdict extraction errors:",
    len(
        verdict_extraction_errors
    ),
)


print()

print(
    "Observed verdicts:",
    sorted(
        observed_verdicts
    ),
)

print(
    "Unexpected verdicts:",
    sorted(
        unexpected_verdicts
    ),
)


print()

print(
    "Verdict counts:"
)

for verdict in sorted(
    allowed_verdicts
):

    print(
        " -",
        verdict,
        ":",
        verdict_counts.get(
            verdict,
            0,
        ),
    )


print()

print(
    "Total extracted verdicts:",
    total_extracted_verdicts,
)


print()

print(
    "Post-review populations:"
)

print(
    " - Accepted:",
    len(
        accepted_seed_slot_ids
    ),
)

print(
    " - Hold Repair:",
    len(
        hold_repair_seed_slot_ids
    ),
)

print(
    " - Hold Relation Review:",
    len(
        hold_relation_review_seed_slot_ids
    ),
)

print(
    " - Rejected:",
    len(
        rejected_seed_slot_ids
    ),
)


print()

print(
    "Validation checks:"
)

for (
    check_name,
    check_value,
) in post_review_preflight_checks.items():

    print(
        " -",
        check_name,
        ":",
        "PASS"
        if check_value
        else "FAIL",
    )


print()


if post_review_preflight_pass:

    print(
        "POST-REVIEW PREFLIGHT: PASS"
    )

else:

    print(
        "POST-REVIEW PREFLIGHT: FAIL"
    )
````

<h2>Post-Review Adjudication — بررسی موارد ردشده</h2>

<p>
مرحله Preflight با موفقیت پایان یافت و مشخص شد که هر 3817 Seed دارای
Semantic Review معتبر است.
از این جمعیت، 3813 Seed با Verdict
<code>ACCEPT_SEED</code>
و 4 Seed با Verdict
<code>REJECT_AS_GOLDEN_SEED</code>
ثبت شده‌اند.
هیچ موردی در وضعیت
<code>HOLD_REPAIR</code>
یا
<code>HOLD_RELATION_REVIEW</code>
وجود ندارد.
</p>

<p>
پیش از ساخت Artifactهای نهایی Golden Dataset v2 باید چهار مورد ردشده
به‌صورت مستقیم بررسی شوند تا مشخص شود تصمیم Semantic Judge بر اساس چه
Dimensions و چه Evidenceای ثبت شده است.
</p>

<p>
در این Step هیچ تصمیم جدیدی توسط کد ساخته نمی‌شود و Verdict مدل نیز
تغییر داده نمی‌شود.
هدف فقط استخراج یک Audit View خوانا از چهار Seed ردشده و کنترل
سازگاری Verdict آن‌ها با Dimensions، دلیل فارسی ثبت‌شده و Evidence
موجود در Review Packet است.
</p>

<p>
برای هر Seed ردشده، شناسه Seed، Claim نماینده، Evidence نماینده،
Dimensions شش‌گانه، دلیل کوتاه فارسی Judge، Review Flags و
Canonical Reason Codes موجود نمایش داده می‌شود.
</p>

<p>
این مرحله هیچ API Call جدیدی انجام نمی‌دهد، هیچ Checkpoint را تغییر
نمی‌دهد و هیچ Artifact نهایی نمی‌نویسد.
</p>

<p>
این Step زمانی PASS است که هر چهار Seed ردشده با Review معتبر و Packet
واقعی قابل بازیابی باشند، Verdict همه آن‌ها دقیقاً
<code>REJECT_AS_GOLDEN_SEED</code>
باشد و اطلاعات لازم برای بررسی تصمیم نهایی بدون Missing Data در دسترس باشد.
</p>

</div>

````python
# Code Cell 17
# REJECTED SEED AUDIT
# READ-ONLY / NO API / NO ARTIFACT WRITE
# =========================================================

required_cell17_names = [
    "rejected_seed_slot_ids",
    "post_review_records_by_seed_id",
    "review_packets_by_slot_id",
]

missing_cell17_names = [
    name
    for name in required_cell17_names
    if name not in globals()
]

assert not missing_cell17_names, (
    "Missing required state: "
    + ", ".join(missing_cell17_names)
)


# =========================================================
# BASIC POPULATION CHECK
# =========================================================

assert len(rejected_seed_slot_ids) == 4, (
    "Expected exactly 4 rejected Seeds from Cell 16, "
    f"found {len(rejected_seed_slot_ids)}."
)


# =========================================================
# BUILD READ-ONLY AUDIT VIEW
# =========================================================

rejected_seed_audit_rows = []

rejected_seed_audit_errors = []


for seed_slot_id in rejected_seed_slot_ids:

    record = post_review_records_by_seed_id.get(
        seed_slot_id
    )

    packet = review_packets_by_slot_id.get(
        seed_slot_id
    )


    if record is None:

        rejected_seed_audit_errors.append({
            "seed_slot_id": seed_slot_id,
            "reason": "MISSING_VALID_REVIEW_RECORD",
        })

        continue


    if packet is None:

        rejected_seed_audit_errors.append({
            "seed_slot_id": seed_slot_id,
            "reason": "MISSING_REVIEW_PACKET",
        })

        continue


    parsed_json = record.get(
        "parsed_json"
    )


    if not isinstance(
        parsed_json,
        dict,
    ):

        rejected_seed_audit_errors.append({
            "seed_slot_id": seed_slot_id,
            "reason": "PARSED_JSON_NOT_OBJECT",
        })

        continue


    verdict = parsed_json.get(
        "verdict"
    )

    dimensions = parsed_json.get(
        "dimensions"
    )

    brief_reason_fa = parsed_json.get(
        "brief_reason_fa"
    )


    if verdict != "REJECT_AS_GOLDEN_SEED":

        rejected_seed_audit_errors.append({
            "seed_slot_id": seed_slot_id,
            "reason": "UNEXPECTED_VERDICT",
            "verdict": verdict,
        })

        continue


    if not isinstance(
        dimensions,
        dict,
    ):

        rejected_seed_audit_errors.append({
            "seed_slot_id": seed_slot_id,
            "reason": "DIMENSIONS_NOT_OBJECT",
        })

        continue


    if not isinstance(
        brief_reason_fa,
        str,
    ) or not brief_reason_fa.strip():

        rejected_seed_audit_errors.append({
            "seed_slot_id": seed_slot_id,
            "reason": "BRIEF_REASON_MISSING",
        })

        continue


    representative_claim = packet.get(
        "representative_claim"
    )

    representative_evidence = packet.get(
        "representative_evidence"
    )

    review_flags = packet.get(
        "review_flags"
    )


    rejected_seed_audit_rows.append({
        "seed_slot_id": seed_slot_id,

        "verdict": verdict,

        "representative_claim": (
            representative_claim
        ),

        "representative_evidence": (
            representative_evidence
        ),

        "dimensions": dimensions,

        "brief_reason_fa": (
            brief_reason_fa.strip()
        ),

        "review_flags": review_flags,

        "canonical_reason_codes": (
            record.get(
                "canonical_reason_codes"
            )
        ),
    })


# =========================================================
# VALIDATION
# =========================================================

cell17_checks = {
    "exactly_4_rejected_seed_ids": (
        len(rejected_seed_slot_ids) == 4
    ),

    "exactly_4_audit_rows": (
        len(rejected_seed_audit_rows) == 4
    ),

    "no_audit_errors": (
        len(rejected_seed_audit_errors) == 0
    ),

    "all_verdicts_are_reject": all(
        row["verdict"]
        == "REJECT_AS_GOLDEN_SEED"
        for row in rejected_seed_audit_rows
    ),

    "all_have_dimensions": all(
        isinstance(
            row["dimensions"],
            dict,
        )
        and bool(
            row["dimensions"]
        )
        for row in rejected_seed_audit_rows
    ),

    "all_have_brief_reason": all(
        isinstance(
            row["brief_reason_fa"],
            str,
        )
        and bool(
            row["brief_reason_fa"].strip()
        )
        for row in rejected_seed_audit_rows
    ),

    "all_have_claim": all(
        row["representative_claim"]
        is not None
        for row in rejected_seed_audit_rows
    ),

    "all_have_evidence": all(
        row["representative_evidence"]
        is not None
        for row in rejected_seed_audit_rows
    ),
}


cell17_pass = all(
    cell17_checks.values()
)


# =========================================================
# OUTPUT
# =========================================================

print(
    "=== POST-REVIEW REJECTED SEED AUDIT ==="
)

print()

print(
    "Rejected Seed population:",
    len(rejected_seed_slot_ids),
)

print(
    "Audit rows built:",
    len(rejected_seed_audit_rows),
)

print(
    "Audit errors:",
    len(rejected_seed_audit_errors),
)


if rejected_seed_audit_errors:

    print()
    print("Audit errors:")

    for error in rejected_seed_audit_errors:

        print(
            json.dumps(
                error,
                ensure_ascii=False,
                indent=2,
            )
        )


print()
print(
    "=============================================="
)


for index, row in enumerate(
    rejected_seed_audit_rows,
    start=1,
):

    print()
    print(
        f"REJECTED SEED {index} / "
        f"{len(rejected_seed_audit_rows)}"
    )

    print(
        "Seed Slot ID:",
        row["seed_slot_id"],
    )

    print(
        "Verdict:",
        row["verdict"],
    )

    print()

    print("Representative Claim:")

    print(
        json.dumps(
            row["representative_claim"],
            ensure_ascii=False,
            indent=2,
        )
    )

    print()

    print("Representative Evidence:")

    print(
        json.dumps(
            row["representative_evidence"],
            ensure_ascii=False,
            indent=2,
        )
    )

    print()

    print("Dimensions:")

    print(
        json.dumps(
            row["dimensions"],
            ensure_ascii=False,
            indent=2,
        )
    )

    print()

    print(
        "Brief Reason FA:",
        row["brief_reason_fa"],
    )

    print()

    print("Review Flags:")

    print(
        json.dumps(
            row["review_flags"],
            ensure_ascii=False,
            indent=2,
        )
    )

    print()

    print("Canonical Reason Codes:")

    print(
        json.dumps(
            row["canonical_reason_codes"],
            ensure_ascii=False,
            indent=2,
        )
    )

    print()

    print(
        "----------------------------------------------"
    )


print()
print(
    "Validation checks:"
)

for (
    check_name,
    check_value,
) in cell17_checks.items():

    print(
        " -",
        check_name,
        ":",
        "PASS"
        if check_value
        else "FAIL",
    )


print()


if cell17_pass:

    print(
        "REJECTED SEED AUDIT: PASS"
    )

else:

    print(
        "REJECTED SEED AUDIT: FAIL"
    )
````

<h2>مرحله ۱۸ — بررسی سازگاری چهار Seed ردشده</h2>

<p>
در مرحله قبل، چهار Seed با رأی
<code>REJECT_AS_GOLDEN_SEED</code>
شناسایی و Audit شدند.
پیش از هرگونه حذف یا پذیرش نهایی، باید سازگاری میان رأی مدل،
ابعاد ساختاری Review، دلیل فارسی و وضعیت مستقل‌بودن Seed بررسی شود.
</p>

<p>
این مرحله صرفاً یک Adjudication محلی و قطعی بر روی داده‌های موجود است.
هیچ فراخوانی API انجام نمی‌شود، هیچ Review قبلی تغییر نمی‌کند،
هیچ Seed حذف نمی‌شود و هیچ Artifact نهایی نوشته نمی‌شود.
</p>

<p>
هدف این مرحله تفکیک Rejectهای درونی‌سازگار از Rejectهایی است که
رأی نهایی آن‌ها با Dimensions ثبت‌شده ناسازگاری دارد و بنابراین
برای تصمیم نهایی نیازمند بررسی جداگانه هستند.
</p>

</div>

````python
# Code Cell 18 — Reject Consistency Adjudication
# =========================================================
# Goals:
# - Audit internal consistency of exactly 4 rejected Seeds
# - Compare verdict against recorded dimensions
# - Detect contradictions between verdict and dimensions
# - Classify each Reject for the NEXT adjudication stage
#
# Safety:
# - ZERO API calls
# - ZERO checkpoint writes
# - ZERO artifact writes
# - ZERO Seed deletion
# - ZERO mutation of prior Review records
# =========================================================


from collections import Counter
import json


# =========================================================
# REQUIRED UPSTREAM STATE
# =========================================================

required_cell18_names = [
    "rejected_seed_slot_ids",
    "post_review_records_by_seed_id",
    "review_packets_by_slot_id",
]


missing_cell18_names = [
    name
    for name in required_cell18_names
    if name not in globals()
]


assert not missing_cell18_names, (
    "Missing required post-review objects: "
    + ", ".join(
        missing_cell18_names
    )
)


assert (
    len(
        rejected_seed_slot_ids
    )
    == 4
), (
    "Expected exactly 4 rejected Seeds."
)


assert (
    len(
        set(
            rejected_seed_slot_ids
        )
    )
    == 4
)


# =========================================================
# EXPECTED REVIEW DIMENSIONS
# =========================================================

required_dimension_names = {
    "standalone_question_possible",
    "answerable_from_attached_evidence",
    "question_scope_unambiguous",
    "independent_evaluation_target",
    "attribution_preservable",
    "requires_external_unprovided_context",
}


# =========================================================
# STRICT ACCEPT-LIKE DIMENSION PROFILE
#
# This does NOT convert a Reject into ACCEPT.
#
# It only asks:
# "If we ignored the verdict string, do the dimensions
# themselves describe a fully acceptable Seed?"
# =========================================================

def dimensions_are_accept_like(
    dimensions,
):

    return (
        dimensions.get(
            "standalone_question_possible"
        )
        is True

        and dimensions.get(
            "answerable_from_attached_evidence"
        )
        is True

        and dimensions.get(
            "question_scope_unambiguous"
        )
        is True

        and dimensions.get(
            "independent_evaluation_target"
        )
        is True

        and dimensions.get(
            "attribution_preservable"
        )
        is True

        and dimensions.get(
            "requires_external_unprovided_context"
        )
        is False
    )


# =========================================================
# DIMENSION FAILURE EXTRACTION
# =========================================================

def derive_dimension_failures(
    dimensions,
):

    failures = []


    if (
        dimensions.get(
            "standalone_question_possible"
        )
        is not True
    ):

        failures.append(
            "STANDALONE_QUESTION_NOT_SUPPORTED"
        )


    if (
        dimensions.get(
            "answerable_from_attached_evidence"
        )
        is not True
    ):

        failures.append(
            "ATTACHED_EVIDENCE_NOT_ANSWERABLE"
        )


    if (
        dimensions.get(
            "question_scope_unambiguous"
        )
        is not True
    ):

        failures.append(
            "QUESTION_SCOPE_NOT_UNAMBIGUOUS"
        )


    if (
        dimensions.get(
            "independent_evaluation_target"
        )
        is not True
    ):

        failures.append(
            "NOT_INDEPENDENT_EVALUATION_TARGET"
        )


    if (
        dimensions.get(
            "attribution_preservable"
        )
        is not True
    ):

        failures.append(
            "ATTRIBUTION_NOT_PRESERVABLE"
        )


    if (
        dimensions.get(
            "requires_external_unprovided_context"
        )
        is not False
    ):

        failures.append(
            "REQUIRES_EXTERNAL_UNPROVIDED_CONTEXT"
        )


    return failures


# =========================================================
# BUILD ADJUDICATION ROWS
# =========================================================

reject_consistency_rows = []

reject_consistency_errors = []


for seed_slot_id in sorted(
    rejected_seed_slot_ids
):

    # -----------------------------------------------------
    # Review record
    # -----------------------------------------------------

    review_record = (
        post_review_records_by_seed_id.get(
            seed_slot_id
        )
    )


    if not isinstance(
        review_record,
        dict,
    ):

        reject_consistency_errors.append({
            "seed_slot_id": (
                seed_slot_id
            ),

            "error": (
                "MISSING_REVIEW_RECORD"
            ),
        })

        continue


    parsed_json = (
        review_record.get(
            "parsed_json"
        )
    )


    if not isinstance(
        parsed_json,
        dict,
    ):

        reject_consistency_errors.append({
            "seed_slot_id": (
                seed_slot_id
            ),

            "error": (
                "MISSING_PARSED_JSON"
            ),
        })

        continue


    verdict = (
        parsed_json.get(
            "verdict"
        )
    )


    if (
        verdict
        != "REJECT_AS_GOLDEN_SEED"
    ):

        reject_consistency_errors.append({
            "seed_slot_id": (
                seed_slot_id
            ),

            "error": (
                "VERDICT_IS_NOT_REJECT"
            ),

            "observed_verdict": (
                verdict
            ),
        })

        continue


    dimensions = (
        parsed_json.get(
            "dimensions"
        )
    )


    if not isinstance(
        dimensions,
        dict,
    ):

        reject_consistency_errors.append({
            "seed_slot_id": (
                seed_slot_id
            ),

            "error": (
                "MISSING_DIMENSIONS"
            ),
        })

        continue


    if (
        set(
            dimensions
        )
        != required_dimension_names
    ):

        reject_consistency_errors.append({
            "seed_slot_id": (
                seed_slot_id
            ),

            "error": (
                "DIMENSION_SCHEMA_MISMATCH"
            ),

            "observed_dimensions": sorted(
                dimensions
            ),
        })

        continue


    # -----------------------------------------------------
    # Review Packet
    # -----------------------------------------------------

    packet = (
        review_packets_by_slot_id.get(
            seed_slot_id
        )
    )


    if not isinstance(
        packet,
        dict,
    ):

        reject_consistency_errors.append({
            "seed_slot_id": (
                seed_slot_id
            ),

            "error": (
                "MISSING_REVIEW_PACKET"
            ),
        })

        continue


    # -----------------------------------------------------
    # Claim / Evidence
    # -----------------------------------------------------

    representative_claim = (
        packet.get(
            "representative_claim"
        )
    )


    representative_evidence = (
        packet.get(
            "representative_evidence"
        )
    )


    if not isinstance(
        representative_claim,
        dict,
    ):

        reject_consistency_errors.append({
            "seed_slot_id": (
                seed_slot_id
            ),

            "error": (
                "MISSING_REPRESENTATIVE_CLAIM"
            ),
        })

        continue


    if not isinstance(
        representative_evidence,
        list,
    ):

        reject_consistency_errors.append({
            "seed_slot_id": (
                seed_slot_id
            ),

            "error": (
                "MISSING_REPRESENTATIVE_EVIDENCE"
            ),
        })

        continue


    claim_text_fa = (
        representative_claim.get(
            "claim_text_fa"
        )
    )


    brief_reason_fa = (
        parsed_json.get(
            "brief_reason_fa"
        )
    )


    review_flags = (
        packet.get(
            "review_flags"
        )
        or []
    )


    canonical_reason_codes = (
        review_record.get(
            "canonical_reason_codes"
        )
        or []
    )


    # -----------------------------------------------------
    # Deterministic consistency analysis
    # -----------------------------------------------------

    accept_like_dimensions = (
        dimensions_are_accept_like(
            dimensions
        )
    )


    dimension_failures = (
        derive_dimension_failures(
            dimensions
        )
    )


    if accept_like_dimensions:

        consistency_class = (
            "REJECT_VERDICT_CONTRADICTS_DIMENSIONS"
        )

        next_action = (
            "MANUAL_REJECT_ADJUDICATION_REQUIRED"
        )

    else:

        consistency_class = (
            "REJECT_VERDICT_SUPPORTED_BY_AT_LEAST_ONE_DIMENSION_FAILURE"
        )

        next_action = (
            "REJECT_EVIDENCE_REVIEW_REQUIRED"
        )


    reject_consistency_rows.append({
        "seed_slot_id": (
            seed_slot_id
        ),

        "verdict": (
            verdict
        ),

        "claim_text_fa": (
            claim_text_fa
        ),

        "brief_reason_fa": (
            brief_reason_fa
        ),

        "dimensions": dict(
            dimensions
        ),

        "accept_like_dimensions": (
            accept_like_dimensions
        ),

        "dimension_failures": (
            dimension_failures
        ),

        "review_flags": list(
            review_flags
        ),

        "canonical_reason_codes": list(
            canonical_reason_codes
        ),

        "consistency_class": (
            consistency_class
        ),

        "next_action": (
            next_action
        ),

        "representative_claim": (
            representative_claim
        ),

        "representative_evidence": (
            representative_evidence
        ),
    })


# =========================================================
# POPULATION VALIDATION
# =========================================================

assert (
    len(
        reject_consistency_rows
    )
    + len(
        reject_consistency_errors
    )
    == 4
)


consistency_class_counts = Counter(
    row[
        "consistency_class"
    ]
    for row in (
        reject_consistency_rows
    )
)


contradictory_reject_seed_slot_ids = sorted(
    row[
        "seed_slot_id"
    ]
    for row in (
        reject_consistency_rows
    )
    if (
        row[
            "accept_like_dimensions"
        ]
        is True
    )
)


dimension_supported_reject_seed_slot_ids = sorted(
    row[
        "seed_slot_id"
    ]
    for row in (
        reject_consistency_rows
    )
    if (
        row[
            "accept_like_dimensions"
        ]
        is False
    )
)


# =========================================================
# VALIDATION CHECKS
# =========================================================

reject_consistency_checks = {
    "exactly_4_rejected_seed_ids": (
        len(
            rejected_seed_slot_ids
        )
        == 4
    ),

    "exactly_4_adjudication_rows": (
        len(
            reject_consistency_rows
        )
        == 4
    ),

    "no_adjudication_errors": (
        len(
            reject_consistency_errors
        )
        == 0
    ),

    "all_verdicts_remain_reject": (
        all(
            row[
                "verdict"
            ]
            == "REJECT_AS_GOLDEN_SEED"
            for row in (
                reject_consistency_rows
            )
        )
    ),

    "all_dimension_schemas_exact": (
        all(
            set(
                row[
                    "dimensions"
                ]
            )
            == required_dimension_names
            for row in (
                reject_consistency_rows
            )
        )
    ),

    "population_partition_complete": (
        (
            len(
                contradictory_reject_seed_slot_ids
            )
            + len(
                dimension_supported_reject_seed_slot_ids
            )
        )
        == 4
    ),

    "partition_is_disjoint": (
        set(
            contradictory_reject_seed_slot_ids
        ).isdisjoint(
            set(
                dimension_supported_reject_seed_slot_ids
            )
        )
    ),
}


reject_consistency_status = (
    "PASS"
    if all(
        reject_consistency_checks.values()
    )
    else "FAIL"
)


# =========================================================
# NEXT READINESS
# =========================================================

if (
    reject_consistency_status
    == "PASS"
):

    reject_adjudication_readiness = (
        "READY_FOR_REJECT_SOURCE_CONTEXT_ADJUDICATION"
    )

else:

    reject_adjudication_readiness = (
        "BLOCKED_REJECT_CONSISTENCY_REPAIR_REQUIRED"
    )


# =========================================================
# OUTPUT
# =========================================================

print(
    "=== POST-REVIEW REJECT CONSISTENCY ADJUDICATION ==="
)

print()

print(
    "Rejected Seed population:",
    len(
        rejected_seed_slot_ids
    ),
)

print(
    "Adjudication rows built:",
    len(
        reject_consistency_rows
    ),
)

print(
    "Adjudication errors:",
    len(
        reject_consistency_errors
    ),
)


print()

print(
    "Consistency class counts:"
)

for (
    consistency_class,
    count,
) in sorted(
    consistency_class_counts.items()
):

    print(
        " -",
        consistency_class,
        ":",
        count,
    )


print()

print(
    "Rejects contradicting their own dimensions:",
    len(
        contradictory_reject_seed_slot_ids
    ),
)

print(
    "Rejects supported by dimension failure:",
    len(
        dimension_supported_reject_seed_slot_ids
    ),
)


print()

for index, row in enumerate(
    reject_consistency_rows,
    start=1,
):

    print(
        "=" * 70
    )

    print()

    print(
        f"REJECT CONSISTENCY {index} / "
        f"{len(reject_consistency_rows)}"
    )

    print(
        "Seed Slot ID:",
        row[
            "seed_slot_id"
        ],
    )

    print(
        "Claim:",
        row[
            "claim_text_fa"
        ],
    )

    print(
        "Judge Reason:",
        row[
            "brief_reason_fa"
        ],
    )

    print(
        "Accept-like Dimensions:",
        row[
            "accept_like_dimensions"
        ],
    )

    print(
        "Dimension Failures:",
        row[
            "dimension_failures"
        ],
    )

    print(
        "Consistency Class:",
        row[
            "consistency_class"
        ],
    )

    print(
        "Next Action:",
        row[
            "next_action"
        ],
    )

    print()


if reject_consistency_errors:

    print(
        "Adjudication errors:"
    )

    for error in (
        reject_consistency_errors
    ):

        print(
            " -",
            error,
        )


print()

print(
    "Validation checks:"
)

for (
    check_name,
    check_value,
) in (
    reject_consistency_checks.items()
):

    print(
        " -",
        check_name,
        ":",
        (
            "PASS"
            if check_value
            else "FAIL"
        ),
    )


print()

print(
    "API calls performed:",
    0,
)

print(
    "Files modified:",
    0,
)

print(
    "Seeds deleted:",
    0,
)

print(
    "Next readiness:",
    reject_adjudication_readiness,
)

print(
    "REJECT CONSISTENCY ADJUDICATION:",
    reject_consistency_status,
)


assert (
    reject_consistency_status
    == "PASS"
), (
    "Reject consistency adjudication failed."
)
````

<h2>مرحله ۱۹ — بازیابی زمینه منبع برای چهار Seed ردشده</h2>

دو Reject دارای حداقل یک Dimension منفی هستند و دو Reject دیگر با
Dimensions ثبت‌شده خود ناسازگاری دارند.

<p>
در این مرحله هنوز هیچ تصمیم نهایی برای حذف، پذیرش یا تعمیر Seed گرفته نمی‌شود.
هدف، بررسی این است که آیا کوتاهی یا ناقص‌بودن Claim ناشی از مرزبندی Claim Extraction
است و اطلاعات لازم در همان سند منبع و پیرامون Evidence وجود دارد، یا اینکه خود منبع نیز
برای ساخت یک هدف ارزیابی مستقل کافی نیست.

این مرحله فقط از داده‌های محلی و Provenance موجود در پروژه استفاده می‌کند.
هیچ API فراخوانی نمی‌شود، هیچ Checkpoint تغییر نمی‌کند و هیچ Artifact نهایی نوشته نمی‌شود.
</p>

````python
# Code Cell 19 — Reject Source Context Discovery
# =========================================================
# Goals:
# - Resolve the exact source document behind all 4 Rejects
# - Discover locally available source/context records
# - Inspect same-document neighboring material
# - Preserve exact provenance
# - Prepare evidence for final Reject adjudication
#
# Safety:
# - ZERO API calls
# - ZERO artifact writes
# - ZERO checkpoint writes
# - ZERO Seed deletion
# - ZERO mutation of prior records
# =========================================================


from pathlib import Path
import json


# =========================================================
# REQUIRED UPSTREAM STATE
# =========================================================

required_cell19_names = [
    "project_root",
    "reject_consistency_rows",
    "rejected_seed_slot_ids",
]


missing_cell19_names = [
    name
    for name in required_cell19_names
    if name not in globals()
]


assert not missing_cell19_names, (
    "Missing required Cell 18 objects: "
    + ", ".join(
        missing_cell19_names
    )
)


assert (
    len(
        reject_consistency_rows
    )
    == 4
)


assert (
    len(
        rejected_seed_slot_ids
    )
    == 4
)


# =========================================================
# SOURCE TARGETS
# =========================================================

reject_source_targets = []


for row in (
    reject_consistency_rows
):

    evidence_items = (
        row[
            "representative_evidence"
        ]
    )


    assert isinstance(
        evidence_items,
        list,
    )


    assert (
        len(
            evidence_items
        )
        >= 1
    )


    primary_evidence = (
        evidence_items[0]
    )


    locator = (
        primary_evidence.get(
            "locator"
        )
        or {}
    )


    source_locator = (
        locator.get(
            "source_locator"
        )
        or {}
    )


    reject_source_targets.append({
        "seed_slot_id": (
            row[
                "seed_slot_id"
            ]
        ),

        "claim_text_fa": (
            row[
                "claim_text_fa"
            ]
        ),

        "consistency_class": (
            row[
                "consistency_class"
            ]
        ),

        "source_id": (
            primary_evidence.get(
                "source_id"
            )
        ),

        "document_id": (
            primary_evidence.get(
                "document_id"
            )
        ),

        "evidence_id": (
            primary_evidence.get(
                "evidence_id"
            )
        ),

        "evidence_quote_exact": (
            primary_evidence.get(
                "evidence_quote_exact"
            )
        ),

        "source_type": (
            primary_evidence.get(
                "source_type"
            )
        ),

        "locator_id": (
            source_locator.get(
                "locator_id"
            )
        ),

        "physical_file_id": (
            source_locator.get(
                "physical_file_id"
            )
        ),

        "program_id": (
            source_locator.get(
                "program_id"
            )
        ),

        "episode_number": (
            source_locator.get(
                "episode_number"
            )
        ),

        "source_line_start": (
            source_locator.get(
                "source_line_start"
            )
        ),

        "source_line_end": (
            source_locator.get(
                "source_line_end"
            )
        ),

        "text_char_start": (
            locator.get(
                "text_char_start"
            )
        ),

        "text_char_end": (
            locator.get(
                "text_char_end"
            )
        ),
    })


assert (
    len(
        reject_source_targets
    )
    == 4
)


# =========================================================
# VALIDATE TARGET PROVENANCE
# =========================================================

source_target_errors = []


for target in (
    reject_source_targets
):

    required_values = [
        "seed_slot_id",
        "source_id",
        "document_id",
        "evidence_id",
        "evidence_quote_exact",
        "physical_file_id",
    ]


    missing_values = [
        name
        for name in required_values
        if not target.get(
            name
        )
    ]


    if missing_values:

        source_target_errors.append({
            "seed_slot_id": (
                target[
                    "seed_slot_id"
                ]
            ),

            "error": (
                "MISSING_SOURCE_PROVENANCE"
            ),

            "missing": (
                missing_values
            ),
        })


# =========================================================
# DISCOVER LOCAL TEXT-LIKE SOURCE FILES
#
# Conservative:
# We inspect existing project files only.
# We do not assume a particular transcript directory.
# =========================================================

excluded_path_parts = {
    ".git",
    ".ipynb_checkpoints",
    "__pycache__",
    ".venv",
    "venv",
    "node_modules",
}


candidate_suffixes = {
    ".json",
    ".jsonl",
    ".txt",
    ".md",
    ".csv",
}


local_source_candidates = []


for path in (
    project_root.rglob("*")
):

    if not path.is_file():
        continue


    if (
        path.suffix.lower()
        not in candidate_suffixes
    ):
        continue


    relative_parts = set(
        path.relative_to(
            project_root
        ).parts
    )


    if (
        relative_parts
        & excluded_path_parts
    ):
        continue


    local_source_candidates.append(
        path
    )


# =========================================================
# SEARCH HELPERS
# =========================================================

def safe_read_text(
    path,
):

    try:

        return path.read_text(
            encoding="utf-8"
        )

    except Exception:

        return None


def normalized_search_values(
    target,
):

    values = [
        target.get(
            "physical_file_id"
        ),
        target.get(
            "document_id"
        ),
        target.get(
            "locator_id"
        ),
        target.get(
            "source_id"
        ),
        target.get(
            "evidence_quote_exact"
        ),
    ]


    return [
        value
        for value in values
        if (
            isinstance(
                value,
                str,
            )
            and value.strip()
        )
    ]


# =========================================================
# DISCOVER FILE MATCHES
# =========================================================

source_discovery_rows = []


for target in (
    reject_source_targets
):

    search_values = (
        normalized_search_values(
            target
        )
    )


    matches = []


    for path in (
        local_source_candidates
    ):

        text = (
            safe_read_text(
                path
            )
        )


        if text is None:
            continue


        matched_values = [
            value
            for value in (
                search_values
            )
            if value in text
        ]


        if not matched_values:
            continue


        matches.append({
            "path": str(
                path.relative_to(
                    project_root
                )
            ),

            "matched_values": (
                matched_values
            ),

            "exact_evidence_quote_found": (
                target[
                    "evidence_quote_exact"
                ]
                in text
            ),
        })


    source_discovery_rows.append({
        **target,

        "candidate_matches": (
            matches
        ),

        "candidate_match_count": (
            len(
                matches
            )
        ),
    })


# =========================================================
# PRIORITIZE SOURCE FILES CONTAINING EXACT QUOTE
# =========================================================

for row in (
    source_discovery_rows
):

    exact_quote_matches = [
        match
        for match in (
            row[
                "candidate_matches"
            ]
        )
        if (
            match[
                "exact_evidence_quote_found"
            ]
        )
    ]


    row[
        "exact_quote_matches"
    ] = (
        exact_quote_matches
    )


    row[
        "exact_quote_match_count"
    ] = (
        len(
            exact_quote_matches
        )
    )


# =========================================================
# EXTRACT LOCAL CONTEXT AROUND EXACT QUOTE
#
# Context is diagnostic only.
# It does NOT replace the frozen Evidence quote.
# =========================================================

CONTEXT_CHAR_RADIUS = 900


for row in (
    source_discovery_rows
):

    context_candidates = []


    for match in (
        row[
            "exact_quote_matches"
        ]
    ):

        path = (
            project_root
            / match[
                "path"
            ]
        )


        text = (
            safe_read_text(
                path
            )
        )


        if text is None:
            continue


        quote = (
            row[
                "evidence_quote_exact"
            ]
        )


        start_index = (
            text.find(
                quote
            )
        )


        if start_index < 0:
            continue


        end_index = (
            start_index
            + len(
                quote
            )
        )


        context_start = max(
            0,
            start_index
            - CONTEXT_CHAR_RADIUS,
        )


        context_end = min(
            len(
                text
            ),
            end_index
            + CONTEXT_CHAR_RADIUS,
        )


        context_text = (
            text[
                context_start:
                context_end
            ]
        )


        context_candidates.append({
            "path": (
                match[
                    "path"
                ]
            ),

            "quote_start_index": (
                start_index
            ),

            "context_start_index": (
                context_start
            ),

            "context_end_index": (
                context_end
            ),

            "context_text": (
                context_text
            ),
        })


    row[
        "context_candidates"
    ] = (
        context_candidates
    )


    row[
        "context_candidate_count"
    ] = (
        len(
            context_candidates
        )
    )


# =========================================================
# DISCOVERY CLASSIFICATION
#
# IMPORTANT:
# This is only source-discovery status.
# It is NOT a Golden adjudication decision.
# =========================================================

for row in (
    source_discovery_rows
):

    if (
        row[
            "context_candidate_count"
        ]
        > 0
    ):

        row[
            "source_context_status"
        ] = (
            "LOCAL_EXACT_CONTEXT_FOUND"
        )

    elif (
        row[
            "candidate_match_count"
        ]
        > 0
    ):

        row[
            "source_context_status"
        ] = (
            "LOCAL_PROVENANCE_MATCH_FOUND_WITHOUT_EXACT_CONTEXT"
        )

    else:

        row[
            "source_context_status"
        ] = (
            "LOCAL_CONTEXT_NOT_DISCOVERED"
        )


# =========================================================
# VALIDATION
# =========================================================

source_context_checks = {
    "exactly_4_source_targets": (
        len(
            reject_source_targets
        )
        == 4
    ),

    "no_source_target_errors": (
        len(
            source_target_errors
        )
        == 0
    ),

    "exactly_4_discovery_rows": (
        len(
            source_discovery_rows
        )
        == 4
    ),

    "all_seed_ids_preserved": (
        {
            row[
                "seed_slot_id"
            ]
            for row in (
                source_discovery_rows
            )
        }
        == set(
            rejected_seed_slot_ids
        )
    ),

    "all_evidence_quotes_preserved": (
        all(
            isinstance(
                row[
                    "evidence_quote_exact"
                ],
                str,
            )
            and bool(
                row[
                    "evidence_quote_exact"
                ].strip()
            )
            for row in (
                source_discovery_rows
            )
        )
    ),
}


source_context_discovery_status = (
    "PASS"
    if all(
        source_context_checks.values()
    )
    else "FAIL"
)


if (
    source_context_discovery_status
    == "PASS"
):

    reject_source_context_readiness = (
        "READY_FOR_CONTEXT_CONTENT_ADJUDICATION"
    )

else:

    reject_source_context_readiness = (
        "BLOCKED_SOURCE_CONTEXT_DISCOVERY_REPAIR_REQUIRED"
    )


# =========================================================
# OUTPUT
# =========================================================

print(
    "=== REJECT SOURCE CONTEXT DISCOVERY ==="
)

print()

print(
    "Rejected Seed targets:",
    len(
        reject_source_targets
    ),
)

print(
    "Local text-like files inspected:",
    len(
        local_source_candidates
    ),
)

print(
    "Source target errors:",
    len(
        source_target_errors
    ),
)


print()


for index, row in enumerate(
    source_discovery_rows,
    start=1,
):

    print(
        "=" * 70
    )

    print()

    print(
        f"REJECT SOURCE {index} / "
        f"{len(source_discovery_rows)}"
    )

    print(
        "Seed Slot ID:",
        row[
            "seed_slot_id"
        ],
    )

    print(
        "Claim:",
        row[
            "claim_text_fa"
        ],
    )

    print(
        "Source ID:",
        row[
            "source_id"
        ],
    )

    print(
        "Document ID:",
        row[
            "document_id"
        ],
    )

    print(
        "Physical File ID:",
        row[
            "physical_file_id"
        ],
    )

    print(
        "Source Lines:",
        row[
            "source_line_start"
        ],
        "-",
        row[
            "source_line_end"
        ],
    )

    print(
        "Consistency Class:",
        row[
            "consistency_class"
        ],
    )

    print(
        "Source Context Status:",
        row[
            "source_context_status"
        ],
    )

    print(
        "Candidate file matches:",
        row[
            "candidate_match_count"
        ],
    )

    print(
        "Exact quote file matches:",
        row[
            "exact_quote_match_count"
        ],
    )

    print(
        "Context candidates:",
        row[
            "context_candidate_count"
        ],
    )


    if (
        row[
            "context_candidates"
        ]
    ):

        print()

        print(
            "FIRST LOCAL CONTEXT:"
        )

        print(
            row[
                "context_candidates"
            ][0][
                "context_text"
            ]
        )


    print()


if source_target_errors:

    print(
        "Source target errors:"
    )

    for error in (
        source_target_errors
    ):

        print(
            " -",
            error,
        )


print()

print(
    "Validation checks:"
)


for (
    check_name,
    check_value,
) in (
    source_context_checks.items()
):

    print(
        " -",
        check_name,
        ":",
        (
            "PASS"
            if check_value
            else "FAIL"
        ),
    )


print()

print(
    "API calls performed:",
    0,
)

print(
    "Files modified:",
    0,
)

print(
    "Seeds deleted:",
    0,
)

print(
    "Next readiness:",
    reject_source_context_readiness,
)

print(
    "REJECT SOURCE CONTEXT DISCOVERY:",
    source_context_discovery_status,
)


assert (
    source_context_discovery_status
    == "PASS"
), (
    "Reject source-context discovery failed."
)
````

<h2>مرحله ۲۰ — انتخاب زمینه واقعی منبع و حذف Matchهای مشتق‌شده</h2>

<p>
مرحله قبل نشان داد که برای هر چهار Seed ردشده، داده محلی مرتبط وجود دارد؛
اما برخی Matchها متعلق به Manifest، Registry یا Artifactهای مشتق‌شده هستند و
Transcript اصلی را نمایش نمی‌دهند.
</p>

<p>
در این مرحله Matchهای موجود رتبه‌بندی می‌شوند تا زمینه‌ای انتخاب شود که
بیشترین تطابق را با Physical File ID، Document ID، Locator و محدوده خطوط
منبع دارد. هدف، جداکردن Source Context واقعی از فایل‌هایی است که صرفاً
Claim یا Evidence را بازنشر کرده‌اند.
</p>

<p>
این مرحله همچنان فقط تشخیصی است. هیچ API فراخوانی نمی‌شود، هیچ Seed
پذیرفته یا حذف نمی‌شود و هیچ Artifact یا Checkpoint قبلی تغییر نمی‌کند.
</p>

</div>

````python
# Code Cell 20 — Exact Source Context Selection
# =========================================================
# Goals:
# - Rank local context candidates conservatively
# - Prefer physical/source transcript evidence
# - Penalize derived Golden/Registry/Review artifacts
# - Extract line-aware context where possible
# - Select one best diagnostic context per rejected Seed
#
# Safety:
# - ZERO API calls
# - ZERO writes
# - ZERO Seed decisions
# =========================================================


from pathlib import Path
import re


# =========================================================
# REQUIRED STATE
# =========================================================

required_cell20_names = [
    "project_root",
    "source_discovery_rows",
    "rejected_seed_slot_ids",
]


missing_cell20_names = [
    name
    for name in required_cell20_names
    if name not in globals()
]


assert not missing_cell20_names, (
    "Missing required Cell 19 objects: "
    + ", ".join(
        missing_cell20_names
    )
)


assert (
    len(
        source_discovery_rows
    )
    == 4
)


# =========================================================
# DERIVED-ARTIFACT PATH SIGNALS
#
# These paths are not forbidden.
# They are merely penalized when choosing the best
# human-readable source context.
# =========================================================

DERIVED_PATH_SIGNALS = [
    "golden_dataset",
    "golden_v2",
    "claim_registry",
    "knowledge_graph",
    "semantic_review",
    "budget_canary",
    "/work/",
    "manifest",
    "report",
    "eligibility",
    "checkpoint",
]


SOURCE_PATH_SIGNALS = [
    "transcript",
    "transcripts",
    "source",
    "sources",
    "raw",
    "corpus",
    "episode",
]


# =========================================================
# TEXT NORMALIZATION
# =========================================================

def normalize_context_text(
    text,
):

    if not isinstance(
        text,
        str,
    ):
        return ""


    return (
        text
        .replace(
            "\r\n",
            "\n",
        )
        .replace(
            "\r",
            "\n",
        )
    )


# =========================================================
# LINE EXTRACTION
#
# Use provenance source-line range when the matched file
# actually has enough physical lines.
# =========================================================

LINE_CONTEXT_RADIUS = 4


def extract_line_context(
    text,
    source_line_start,
    source_line_end,
):

    if not isinstance(
        source_line_start,
        int,
    ):
        return None


    if not isinstance(
        source_line_end,
        int,
    ):
        return None


    normalized = (
        normalize_context_text(
            text
        )
    )


    lines = (
        normalized.splitlines()
    )


    if not lines:
        return None


    if (
        source_line_start < 1
        or source_line_end < source_line_start
        or source_line_start > len(
            lines
        )
    ):
        return None


    start_index = max(
        0,
        source_line_start
        - 1
        - LINE_CONTEXT_RADIUS,
    )


    end_index = min(
        len(
            lines
        ),
        source_line_end
        + LINE_CONTEXT_RADIUS,
    )


    selected_lines = []


    for line_index in range(
        start_index,
        end_index,
    ):

        selected_lines.append(
            f"{line_index + 1}: "
            f"{lines[line_index]}"
        )


    return {
        "line_start": (
            start_index + 1
        ),

        "line_end": (
            end_index
        ),

        "text": (
            "\n".join(
                selected_lines
            )
        ),
    }


# =========================================================
# QUOTE-CENTERED FALLBACK
# =========================================================

QUOTE_CONTEXT_RADIUS = 1200


def extract_quote_context(
    text,
    quote,
):

    normalized = (
        normalize_context_text(
            text
        )
    )


    if not isinstance(
        quote,
        str,
    ):
        return None


    quote = quote.strip()


    if not quote:
        return None


    index = (
        normalized.find(
            quote
        )
    )


    if index < 0:

        # Conservative punctuation fallback.
        relaxed_quote = (
            quote.rstrip(
                "،,.؛;:!?؟ "
            )
        )


        if not relaxed_quote:
            return None


        index = (
            normalized.find(
                relaxed_quote
            )
        )


        if index < 0:
            return None


        quote = (
            relaxed_quote
        )


    start = max(
        0,
        index
        - QUOTE_CONTEXT_RADIUS,
    )


    end = min(
        len(
            normalized
        ),
        index
        + len(
            quote
        )
        + QUOTE_CONTEXT_RADIUS,
    )


    return {
        "char_start": (
            start
        ),

        "char_end": (
            end
        ),

        "text": (
            normalized[
                start:end
            ]
        ),
    }


# =========================================================
# PATH CLASSIFICATION
# =========================================================

def path_signal_profile(
    relative_path,
):

    lower_path = (
        str(
            relative_path
        )
        .lower()
        .replace(
            "\\",
            "/",
        )
    )


    derived_hits = [
        signal
        for signal in (
            DERIVED_PATH_SIGNALS
        )
        if signal in lower_path
    ]


    source_hits = [
        signal
        for signal in (
            SOURCE_PATH_SIGNALS
        )
        if signal in lower_path
    ]


    return (
        derived_hits,
        source_hits,
    )


# =========================================================
# SCORE CANDIDATE
#
# Score is only for context discovery.
# It does NOT determine Golden acceptance.
# =========================================================

def score_source_candidate(
    *,
    row,
    path,
    text,
):

    score = 0

    reasons = []


    normalized_text = (
        normalize_context_text(
            text
        )
    )


    derived_hits, source_hits = (
        path_signal_profile(
            path
        )
    )


    # -----------------------------------------------------
    # Exact Evidence quote
    # -----------------------------------------------------

    quote = (
        row[
            "evidence_quote_exact"
        ]
    )


    exact_quote_found = (
        isinstance(
            quote,
            str,
        )
        and quote in normalized_text
    )


    if exact_quote_found:

        score += 30

        reasons.append(
            "EXACT_EVIDENCE_QUOTE"
        )


    # -----------------------------------------------------
    # Physical file identity
    # -----------------------------------------------------

    physical_file_id = (
        row.get(
            "physical_file_id"
        )
    )


    physical_id_found = (
        isinstance(
            physical_file_id,
            str,
        )
        and physical_file_id
        in normalized_text
    )


    if physical_id_found:

        score += 12

        reasons.append(
            "PHYSICAL_FILE_ID"
        )


    # -----------------------------------------------------
    # Document identity
    # -----------------------------------------------------

    document_id = (
        row.get(
            "document_id"
        )
    )


    document_id_found = (
        isinstance(
            document_id,
            str,
        )
        and document_id
        in normalized_text
    )


    if document_id_found:

        score += 8

        reasons.append(
            "DOCUMENT_ID"
        )


    # -----------------------------------------------------
    # Locator identity
    # -----------------------------------------------------

    locator_id = (
        row.get(
            "locator_id"
        )
    )


    locator_id_found = (
        isinstance(
            locator_id,
            str,
        )
        and locator_id
        in normalized_text
    )


    if locator_id_found:

        score += 8

        reasons.append(
            "LOCATOR_ID"
        )


    # -----------------------------------------------------
    # Source path hints
    # -----------------------------------------------------

    if source_hits:

        score += (
            6
            * len(
                source_hits
            )
        )

        reasons.extend(
            f"SOURCE_PATH:{hit}"
            for hit in (
                source_hits
            )
        )


    # -----------------------------------------------------
    # Derived artifact penalty
    # -----------------------------------------------------

    if derived_hits:

        score -= (
            15
            * len(
                derived_hits
            )
        )

        reasons.extend(
            f"DERIVED_PATH:{hit}"
            for hit in (
                derived_hits
            )
        )


    # -----------------------------------------------------
    # Strong penalty when the context visibly looks like
    # serialized Golden/Registry metadata.
    # -----------------------------------------------------

    serialized_metadata_markers = [
        '"claim_id"',
        '"eligibility_policy_version"',
        '"final_seed_decision"',
        '"review_flags"',
        '"structural_eligibility"',
        '"manifest_version"',
    ]


    serialized_hits = sum(
        marker in normalized_text
        for marker in (
            serialized_metadata_markers
        )
    )


    if serialized_hits >= 3:

        score -= 35

        reasons.append(
            "SERIALIZED_DERIVED_METADATA"
        )


    # -----------------------------------------------------
    # Line-aware source compatibility
    # -----------------------------------------------------

    line_context = (
        extract_line_context(
            normalized_text,
            row.get(
                "source_line_start"
            ),
            row.get(
                "source_line_end"
            ),
        )
    )


    line_context_contains_quote = (
        line_context is not None
        and isinstance(
            quote,
            str,
        )
        and quote.rstrip(
            "،,.؛;:!?؟ "
        )
        in (
            line_context[
                "text"
            ]
        )
    )


    if line_context_contains_quote:

        score += 50

        reasons.append(
            "PROVENANCE_LINE_RANGE_CONTAINS_QUOTE"
        )


    return {
        "score": (
            score
        ),

        "score_reasons": (
            reasons
        ),

        "derived_path_hits": (
            derived_hits
        ),

        "source_path_hits": (
            source_hits
        ),

        "exact_quote_found": (
            exact_quote_found
        ),

        "physical_id_found": (
            physical_id_found
        ),

        "document_id_found": (
            document_id_found
        ),

        "locator_id_found": (
            locator_id_found
        ),

        "line_context": (
            line_context
        ),

        "line_context_contains_quote": (
            line_context_contains_quote
        ),
    }


# =========================================================
# RE-SCORE ALL MATCHED FILES
# =========================================================

exact_source_selection_rows = []

source_selection_errors = []


for row in (
    source_discovery_rows
):

    ranked_candidates = []


    candidate_paths = sorted({
        match[
            "path"
        ]
        for match in (
            row[
                "candidate_matches"
            ]
        )
    })


    for relative_path in (
        candidate_paths
    ):

        path = (
            project_root
            / relative_path
        )


        try:

            text = (
                path.read_text(
                    encoding="utf-8"
                )
            )

        except Exception:

            continue


        score_result = (
            score_source_candidate(
                row=row,
                path=relative_path,
                text=text,
            )
        )


        quote_context = (
            extract_quote_context(
                text,
                row[
                    "evidence_quote_exact"
                ],
            )
        )


        # Prefer provenance-line context only when the
        # quote is actually present inside that line range.
        if (
            score_result[
                "line_context_contains_quote"
            ]
        ):

            selected_context_kind = (
                "PROVENANCE_LINE_CONTEXT"
            )

            selected_context_text = (
                score_result[
                    "line_context"
                ][
                    "text"
                ]
            )

        elif (
            quote_context
            is not None
        ):

            selected_context_kind = (
                "QUOTE_CENTERED_CONTEXT"
            )

            selected_context_text = (
                quote_context[
                    "text"
                ]
            )

        else:

            selected_context_kind = (
                "NO_USABLE_CONTEXT"
            )

            selected_context_text = (
                None
            )


        ranked_candidates.append({
            "path": (
                relative_path
            ),

            **score_result,

            "selected_context_kind": (
                selected_context_kind
            ),

            "selected_context_text": (
                selected_context_text
            ),
        })


    ranked_candidates.sort(
        key=lambda item: (
            item[
                "score"
            ],

            item[
                "line_context_contains_quote"
            ],

            item[
                "exact_quote_found"
            ],

            -len(
                item[
                    "derived_path_hits"
                ]
            ),

            item[
                "path"
            ],
        ),
        reverse=True,
    )


    usable_candidates = [
        candidate
        for candidate in (
            ranked_candidates
        )
        if (
            candidate[
                "selected_context_text"
            ]
        )
    ]


    if usable_candidates:

        best_candidate = (
            usable_candidates[0]
        )

    else:

        best_candidate = None


    if best_candidate is None:

        source_selection_errors.append({
            "seed_slot_id": (
                row[
                    "seed_slot_id"
                ]
            ),

            "error": (
                "NO_USABLE_LOCAL_CONTEXT"
            ),
        })


    exact_source_selection_rows.append({
        "seed_slot_id": (
            row[
                "seed_slot_id"
            ]
        ),

        "claim_text_fa": (
            row[
                "claim_text_fa"
            ]
        ),

        "consistency_class": (
            row[
                "consistency_class"
            ]
        ),

        "source_id": (
            row[
                "source_id"
            ]
        ),

        "document_id": (
            row[
                "document_id"
            ]
        ),

        "physical_file_id": (
            row[
                "physical_file_id"
            ]
        ),

        "evidence_quote_exact": (
            row[
                "evidence_quote_exact"
            ]
        ),

        "source_line_start": (
            row[
                "source_line_start"
            ]
        ),

        "source_line_end": (
            row[
                "source_line_end"
            ]
        ),

        "ranked_candidate_count": (
            len(
                ranked_candidates
            )
        ),

        "best_candidate": (
            best_candidate
        ),

        "top_candidates": (
            ranked_candidates[
                :5
            ]
        ),
    })


# =========================================================
# VALIDATION
# =========================================================

exact_source_selection_checks = {
    "exactly_4_selection_rows": (
        len(
            exact_source_selection_rows
        )
        == 4
    ),

    "all_seed_ids_preserved": (
        {
            row[
                "seed_slot_id"
            ]
            for row in (
                exact_source_selection_rows
            )
        }
        == set(
            rejected_seed_slot_ids
        )
    ),

    "all_have_best_candidate": (
        all(
            row[
                "best_candidate"
            ]
            is not None
            for row in (
                exact_source_selection_rows
            )
        )
    ),

    "all_best_candidates_have_context": (
        all(
            isinstance(
                row[
                    "best_candidate"
                ][
                    "selected_context_text"
                ],
                str,
            )
            and bool(
                row[
                    "best_candidate"
                ][
                    "selected_context_text"
                ].strip()
            )
            for row in (
                exact_source_selection_rows
            )
            if (
                row[
                    "best_candidate"
                ]
                is not None
            )
        )
    ),

    "no_selection_errors": (
        len(
            source_selection_errors
        )
        == 0
    ),
}


exact_source_selection_status = (
    "PASS"
    if all(
        exact_source_selection_checks.values()
    )
    else "FAIL"
)


if (
    exact_source_selection_status
    == "PASS"
):

    exact_source_selection_readiness = (
        "READY_FOR_FINAL_REJECT_CONTENT_ADJUDICATION"
    )

else:

    exact_source_selection_readiness = (
        "BLOCKED_SOURCE_SELECTION_REPAIR_REQUIRED"
    )


# =========================================================
# OUTPUT
# =========================================================

print(
    "=== EXACT REJECT SOURCE CONTEXT SELECTION ==="
)

print()

print(
    "Selection rows:",
    len(
        exact_source_selection_rows
    ),
)

print(
    "Selection errors:",
    len(
        source_selection_errors
    ),
)


print()


for index, row in enumerate(
    exact_source_selection_rows,
    start=1,
):

    print(
        "=" * 70
    )

    print()

    print(
        f"EXACT SOURCE {index} / "
        f"{len(exact_source_selection_rows)}"
    )

    print(
        "Seed Slot ID:",
        row[
            "seed_slot_id"
        ],
    )

    print(
        "Claim:",
        row[
            "claim_text_fa"
        ],
    )

    print(
        "Evidence Quote:",
        row[
            "evidence_quote_exact"
        ],
    )


    best = (
        row[
            "best_candidate"
        ]
    )


    if best is not None:

        print(
            "Selected Path:",
            best[
                "path"
            ],
        )

        print(
            "Selection Score:",
            best[
                "score"
            ],
        )

        print(
            "Context Kind:",
            best[
                "selected_context_kind"
            ],
        )

        print(
            "Line-range quote match:",
            best[
                "line_context_contains_quote"
            ],
        )

        print(
            "Derived-path signals:",
            best[
                "derived_path_hits"
            ],
        )

        print(
            "Source-path signals:",
            best[
                "source_path_hits"
            ],
        )

        print(
            "Selection Reasons:",
            best[
                "score_reasons"
            ],
        )

        print()

        print(
            "SELECTED SOURCE CONTEXT:"
        )

        print(
            best[
                "selected_context_text"
            ]
        )


    print()

    print(
        "TOP CANDIDATE PATHS:"
    )


    for candidate in (
        row[
            "top_candidates"
        ]
    ):

        print(
            " -",
            candidate[
                "score"
            ],
            "|",
            candidate[
                "selected_context_kind"
            ],
            "|",
            candidate[
                "path"
            ],
        )


    print()


print(
    "Validation checks:"
)


for (
    check_name,
    check_value,
) in (
    exact_source_selection_checks.items()
):

    print(
        " -",
        check_name,
        ":",
        (
            "PASS"
            if check_value
            else "FAIL"
        ),
    )


print()

print(
    "API calls performed:",
    0,
)

print(
    "Files modified:",
    0,
)

print(
    "Seeds deleted:",
    0,
)

print(
    "Next readiness:",
    exact_source_selection_readiness,
)

print(
    "EXACT SOURCE CONTEXT SELECTION:",
    exact_source_selection_status,
)


assert (
    exact_source_selection_status
    == "PASS"
), (
    "Exact source-context selection failed."
)
````

````python
# Code Cell 21 — Final Reject Content Adjudication
# =========================================================
# Goals:
# - Adjudicate the 4 rejected Seed Slots conservatively
# - Distinguish CURRENT-SEED rejection from SOURCE knowledge loss
# - Preserve repairable source knowledge for later re-extraction
# - Produce deterministic post-review decisions
#
# IMPORTANT:
# - We do NOT expand frozen Evidence in place.
# - We do NOT silently turn source context into Seed evidence.
# - A rejected current Seed may still become a repair candidate.
#
# Safety:
# - ZERO API calls
# - ZERO artifact writes
# - ZERO checkpoint writes
# - ZERO mutation of prior Review records
# =========================================================


from collections import Counter


# =========================================================
# REQUIRED STATE
# =========================================================

required_cell21_names = [
    "exact_source_selection_rows",
    "reject_consistency_rows",
    "rejected_seed_slot_ids",
]


missing_cell21_names = [
    name
    for name in required_cell21_names
    if name not in globals()
]


assert not missing_cell21_names, (
    "Missing required adjudication objects: "
    + ", ".join(
        missing_cell21_names
    )
)


assert (
    len(
        exact_source_selection_rows
    )
    == 4
)


assert (
    len(
        reject_consistency_rows
    )
    == 4
)


assert (
    len(
        rejected_seed_slot_ids
    )
    == 4
)


# =========================================================
# INDEX INPUTS
# =========================================================

consistency_by_seed_id = {
    row[
        "seed_slot_id"
    ]: row
    for row in (
        reject_consistency_rows
    )
}


source_selection_by_seed_id = {
    row[
        "seed_slot_id"
    ]: row
    for row in (
        exact_source_selection_rows
    )
}


assert (
    set(
        consistency_by_seed_id
    )
    == set(
        rejected_seed_slot_ids
    )
)


assert (
    set(
        source_selection_by_seed_id
    )
    == set(
        rejected_seed_slot_ids
    )
)


# =========================================================
# FROZEN ADJUDICATION POLICY
#
# This policy is intentionally explicit for the 4 observed
# Rejects. It does not invent a new Semantic Judge verdict.
#
# CURRENT_SEED_REJECT:
#   The current frozen Seed is not admitted to Golden v2.
#
# REPAIR_FROM_SOURCE:
#   Current Seed remains rejected, but source context proves
#   that useful knowledge exists and should be re-extracted
#   as a NEW candidate rather than modifying frozen evidence.
#
# REJECT_NO_REPAIR:
#   Current Seed remains rejected and no automatic repair is
#   authorized by this adjudication.
# =========================================================

FINAL_REJECT_ADJUDICATION_POLICY_VERSION = (
    "golden_v2_reject_content_adjudication_v1"
)


expected_reject_adjudications = {

    # -----------------------------------------------------
    # 1
    #
    # Frozen claim:
    # "امنیت شغلی موضوع درونیه"
    #
    # Transcript immediately explains the intended relation:
    # internal job security depends on alignment between
    # one's abilities/skills and market needs.
    #
    # Current Seed evidence itself was judged not answerable,
    # so it remains rejected. Source knowledge is repairable.
    # -----------------------------------------------------

    "gsv2_slot_180fd2a592d1442a829f39d9": {
        "current_seed_decision": (
            "REJECT_CURRENT_SEED"
        ),

        "source_knowledge_decision": (
            "REPAIR_FROM_SOURCE"
        ),

        "repair_mode": (
            "REEXTRACT_COMPLETE_CLAIM_WITH_SOURCE_CONTEXT"
        ),

        "adjudication_reason": (
            "CURRENT_EVIDENCE_INCOMPLETE_BUT_SOURCE_CONTEXT_"
            "SUPPORTS_COMPLETE_RELATION"
        ),
    },


    # -----------------------------------------------------
    # 2
    #
    # Frozen claim:
    # "افزایش پول تو جیبی می‌کنن،"
    #
    # The fragment is incomplete as a standalone Seed.
    # Transcript continues into the actual question and
    # answer about proportional increases.
    #
    # Do not override the Reject merely because dimensions
    # were internally contradictory.
    # -----------------------------------------------------

    "gsv2_slot_a3d0743900628e5199e72bdc": {
        "current_seed_decision": (
            "REJECT_CURRENT_SEED"
        ),

        "source_knowledge_decision": (
            "REPAIR_FROM_SOURCE"
        ),

        "repair_mode": (
            "REEXTRACT_COMPLETE_CLAIM_WITH_SOURCE_CONTEXT"
        ),

        "adjudication_reason": (
            "FRAGMENTARY_CURRENT_CLAIM_WITH_RECOVERABLE_"
            "SOURCE_PROPOSITION"
        ),
    },


    # -----------------------------------------------------
    # 3
    #
    # Frozen claim:
    # "تبصره اینه که در این شرایط نگیریم"
    #
    # In isolation, the referents "تبصره" and "این شرایط"
    # are unresolved. Transcript provides the debt/borrowing
    # rule and its exceptions.
    #
    # Current Seed remains rejected; knowledge is repairable.
    # -----------------------------------------------------

    "gsv2_slot_dda80211d464c74c787be5d1": {
        "current_seed_decision": (
            "REJECT_CURRENT_SEED"
        ),

        "source_knowledge_decision": (
            "REPAIR_FROM_SOURCE"
        ),

        "repair_mode": (
            "REEXTRACT_COMPLETE_CLAIM_WITH_SOURCE_CONTEXT"
        ),

        "adjudication_reason": (
            "UNRESOLVED_REFERENTS_IN_CURRENT_CLAIM_WITH_"
            "RECOVERABLE_SOURCE_RULE"
        ),
    },


    # -----------------------------------------------------
    # 4
    #
    # Frozen claim:
    # "دارم می‌رم توی سطح اقتصادی جدید"
    #
    # The isolated sentence is ambiguous, but transcript
    # explains it as a reason borrowing may be undesirable:
    # spending beyond one's actual economic level.
    #
    # Current Seed remains rejected; source knowledge is
    # preserved for clean re-extraction.
    # -----------------------------------------------------

    "gsv2_slot_df894995039d3f3feb356407": {
        "current_seed_decision": (
            "REJECT_CURRENT_SEED"
        ),

        "source_knowledge_decision": (
            "REPAIR_FROM_SOURCE"
        ),

        "repair_mode": (
            "REEXTRACT_COMPLETE_CLAIM_WITH_SOURCE_CONTEXT"
        ),

        "adjudication_reason": (
            "AMBIGUOUS_CURRENT_FRAGMENT_WITH_RECOVERABLE_"
            "SOURCE_RELATION"
        ),
    },
}


# =========================================================
# ENSURE POLICY MATCHES EXACT OBSERVED POPULATION
# =========================================================

assert (
    set(
        expected_reject_adjudications
    )
    == set(
        rejected_seed_slot_ids
    )
), (
    "Reject adjudication policy does not match "
    "the exact observed Reject population."
)


# =========================================================
# BUILD FINAL ADJUDICATION ROWS
# =========================================================

final_reject_adjudication_rows = []

final_reject_adjudication_errors = []


for seed_slot_id in sorted(
    rejected_seed_slot_ids
):

    consistency_row = (
        consistency_by_seed_id[
            seed_slot_id
        ]
    )


    source_row = (
        source_selection_by_seed_id[
            seed_slot_id
        ]
    )


    policy = (
        expected_reject_adjudications[
            seed_slot_id
        ]
    )


    best_candidate = (
        source_row.get(
            "best_candidate"
        )
    )


    if not isinstance(
        best_candidate,
        dict,
    ):

        final_reject_adjudication_errors.append({
            "seed_slot_id": (
                seed_slot_id
            ),

            "error": (
                "MISSING_SELECTED_SOURCE_CONTEXT"
            ),
        })

        continue


    selected_context_text = (
        best_candidate.get(
            "selected_context_text"
        )
    )


    if not (
        isinstance(
            selected_context_text,
            str,
        )
        and selected_context_text.strip()
    ):

        final_reject_adjudication_errors.append({
            "seed_slot_id": (
                seed_slot_id
            ),

            "error": (
                "EMPTY_SELECTED_SOURCE_CONTEXT"
            ),
        })

        continue


    final_reject_adjudication_rows.append({
        "seed_slot_id": (
            seed_slot_id
        ),

        "original_semantic_verdict": (
            consistency_row[
                "verdict"
            ]
        ),

        "claim_text_fa": (
            consistency_row[
                "claim_text_fa"
            ]
        ),

        "judge_reason_fa": (
            consistency_row[
                "brief_reason_fa"
            ]
        ),

        "semantic_dimensions": dict(
            consistency_row[
                "dimensions"
            ]
        ),

        "semantic_consistency_class": (
            consistency_row[
                "consistency_class"
            ]
        ),

        "selected_source_path": (
            best_candidate[
                "path"
            ]
        ),

        "selected_context_kind": (
            best_candidate[
                "selected_context_kind"
            ]
        ),

        "selected_source_context": (
            selected_context_text
        ),

        "current_seed_decision": (
            policy[
                "current_seed_decision"
            ]
        ),

        "source_knowledge_decision": (
            policy[
                "source_knowledge_decision"
            ]
        ),

        "repair_mode": (
            policy[
                "repair_mode"
            ]
        ),

        "adjudication_reason": (
            policy[
                "adjudication_reason"
            ]
        ),

        "policy_version": (
            FINAL_REJECT_ADJUDICATION_POLICY_VERSION
        ),
    })


# =========================================================
# DECISION POPULATIONS
# =========================================================

current_seed_decision_counts = Counter(
    row[
        "current_seed_decision"
    ]
    for row in (
        final_reject_adjudication_rows
    )
)


source_knowledge_decision_counts = Counter(
    row[
        "source_knowledge_decision"
    ]
    for row in (
        final_reject_adjudication_rows
    )
)


final_rejected_seed_slot_ids = sorted(
    row[
        "seed_slot_id"
    ]
    for row in (
        final_reject_adjudication_rows
    )
    if (
        row[
            "current_seed_decision"
        ]
        == "REJECT_CURRENT_SEED"
    )
)


source_repair_seed_slot_ids = sorted(
    row[
        "seed_slot_id"
    ]
    for row in (
        final_reject_adjudication_rows
    )
    if (
        row[
            "source_knowledge_decision"
        ]
        == "REPAIR_FROM_SOURCE"
    )
)


# =========================================================
# CRITICAL SEMANTIC DISTINCTION
#
# Accepted Golden Seed population remains 3813.
# The 4 rejected current Seed Slots do NOT become accepted.
#
# Repair candidates are a separate future population.
# =========================================================

post_adjudication_accepted_seed_slot_ids = sorted(
    set(
        accepted_seed_slot_ids
    )
)


assert (
    len(
        post_adjudication_accepted_seed_slot_ids
    )
    == 3813
)


assert (
    set(
        post_adjudication_accepted_seed_slot_ids
    ).isdisjoint(
        set(
            final_rejected_seed_slot_ids
        )
    )
)


assert (
    set(
        source_repair_seed_slot_ids
    )
    == set(
        final_rejected_seed_slot_ids
    )
)


# =========================================================
# VALIDATION CHECKS
# =========================================================

final_reject_adjudication_checks = {
    "exactly_4_adjudication_rows": (
        len(
            final_reject_adjudication_rows
        )
        == 4
    ),

    "no_adjudication_errors": (
        len(
            final_reject_adjudication_errors
        )
        == 0
    ),

    "all_original_verdicts_are_reject": (
        all(
            row[
                "original_semantic_verdict"
            ]
            == "REJECT_AS_GOLDEN_SEED"
            for row in (
                final_reject_adjudication_rows
            )
        )
    ),

    "all_current_seeds_remain_rejected": (
        len(
            final_rejected_seed_slot_ids
        )
        == 4
    ),

    "all_source_knowledge_preserved_for_repair": (
        len(
            source_repair_seed_slot_ids
        )
        == 4
    ),

    "accepted_population_remains_3813": (
        len(
            post_adjudication_accepted_seed_slot_ids
        )
        == 3813
    ),

    "accepted_and_rejected_are_disjoint": (
        set(
            post_adjudication_accepted_seed_slot_ids
        ).isdisjoint(
            set(
                final_rejected_seed_slot_ids
            )
        )
    ),

    "full_population_accounted_for": (
        (
            len(
                post_adjudication_accepted_seed_slot_ids
            )
            + len(
                final_rejected_seed_slot_ids
            )
        )
        == 3817
    ),

    "all_selected_contexts_are_provenance_line_context": (
        all(
            row[
                "selected_context_kind"
            ]
            == "PROVENANCE_LINE_CONTEXT"
            for row in (
                final_reject_adjudication_rows
            )
        )
    ),
}


final_reject_adjudication_status = (
    "PASS"
    if all(
        final_reject_adjudication_checks.values()
    )
    else "FAIL"
)


if (
    final_reject_adjudication_status
    == "PASS"
):

    golden_v2_post_review_readiness = (
        "READY_FOR_POST_REVIEW_DECISION_ARTIFACTS"
    )

else:

    golden_v2_post_review_readiness = (
        "BLOCKED_REJECT_ADJUDICATION_REPAIR_REQUIRED"
    )


# =========================================================
# OUTPUT
# =========================================================

print(
    "=== FINAL REJECT CONTENT ADJUDICATION ==="
)

print()

print(
    "Adjudication policy:",
    FINAL_REJECT_ADJUDICATION_POLICY_VERSION,
)

print(
    "Adjudication rows:",
    len(
        final_reject_adjudication_rows
    ),
)

print(
    "Adjudication errors:",
    len(
        final_reject_adjudication_errors
    ),
)


print()

print(
    "Current Seed decisions:",
    dict(
        sorted(
            current_seed_decision_counts.items()
        )
    ),
)

print(
    "Source knowledge decisions:",
    dict(
        sorted(
            source_knowledge_decision_counts.items()
        )
    ),
)


print()


for index, row in enumerate(
    final_reject_adjudication_rows,
    start=1,
):

    print(
        "=" * 70
    )

    print()

    print(
        f"FINAL REJECT {index} / "
        f"{len(final_reject_adjudication_rows)}"
    )

    print(
        "Seed Slot ID:",
        row[
            "seed_slot_id"
        ],
    )

    print(
        "Claim:",
        row[
            "claim_text_fa"
        ],
    )

    print(
        "Semantic consistency:",
        row[
            "semantic_consistency_class"
        ],
    )

    print(
        "Current Seed decision:",
        row[
            "current_seed_decision"
        ],
    )

    print(
        "Source knowledge decision:",
        row[
            "source_knowledge_decision"
        ],
    )

    print(
        "Repair mode:",
        row[
            "repair_mode"
        ],
    )

    print(
        "Adjudication reason:",
        row[
            "adjudication_reason"
        ],
    )

    print(
        "Source:",
        row[
            "selected_source_path"
        ],
    )

    print()


print(
    "Final populations:"
)

print(
    " - Accepted current Seeds:",
    len(
        post_adjudication_accepted_seed_slot_ids
    ),
)

print(
    " - Rejected current Seeds:",
    len(
        final_rejected_seed_slot_ids
    ),
)

print(
    " - Source repair candidates:",
    len(
        source_repair_seed_slot_ids
    ),
)

print(
    " - Total accounted:",
    (
        len(
            post_adjudication_accepted_seed_slot_ids
        )
        + len(
            final_rejected_seed_slot_ids
        )
    ),
)


print()

print(
    "Validation checks:"
)


for (
    check_name,
    check_value,
) in (
    final_reject_adjudication_checks.items()
):

    print(
        " -",
        check_name,
        ":",
        (
            "PASS"
            if check_value
            else "FAIL"
        ),
    )


print()

print(
    "API calls performed:",
    0,
)

print(
    "Files modified:",
    0,
)

print(
    "Prior Semantic Reviews modified:",
    0,
)

print(
    "Accepted Seeds silently expanded with source context:",
    0,
)

print(
    "Next readiness:",
    golden_v2_post_review_readiness,
)

print(
    "FINAL REJECT CONTENT ADJUDICATION:",
    final_reject_adjudication_status,
)


assert (
    final_reject_adjudication_status
    == "PASS"
), (
    "Final Reject content adjudication failed."
)
````

<h2>مرحله ۲۲ — ساخت Artifactهای رسمی تصمیم Post-Review</h2>

<p>
پس از تکمیل Semantic Review و Adjudication چهار Reject، اکنون تصمیم نهایی
سطح Seed برای تمام ۳۸۱۷ Seed مشخص است. تعداد ۳۸۱۳ Seed برای ادامه مسیر
Golden v2 پذیرفته شده‌اند و ۴ Seed فعلی رد شده‌اند.
</p>

<p>
چهار Seed ردشده از Golden Seed Pool جاری خارج می‌شوند، اما Provenance و
اطلاعات Repair آن‌ها به‌طور مستقل حفظ می‌شود تا دانش موجود در Transcript
در مرحله Re-extraction از بین نرود.
</p>

<p>
در این مرحله Artifactهای رسمی Post-Review به‌صورت Deterministic ساخته و
Freeze می‌شوند. این Artifactها نتیجه تصمیم سطح Seed هستند و نباید با
Golden Dataset نهایی سؤال/جواب اشتباه گرفته شوند.
</p>

<p>
هیچ API فراخوانی نمی‌شود و Semantic Reviewهای قبلی نیز تغییر نمی‌کنند.
اگر Artifact هم‌نام از قبل وجود داشته باشد، فقط در صورت تطابق دقیق محتوا
قابل استفاده است و محتوای متفاوت روی فایل Frozen نوشته نخواهد شد.
</p>

</div>

````python
# Code Cell 22 — Post-Review Decision Artifacts
# =========================================================
# Goals:
# - Freeze all 3817 final Seed-level judgments
# - Freeze the 3813 accepted current Seed pool
# - Freeze a deterministic Post-Review summary report
# - Preserve 4 rejected Seeds as separate repair candidates
#
# IMPORTANT:
# These are Seed-level decision artifacts.
# They are NOT yet the final question/answer Golden Dataset.
#
# Safety:
# - ZERO API calls
# - ZERO Semantic Review mutation
# - ZERO overwrite of different frozen content
# =========================================================


from pathlib import Path
from collections import Counter
import hashlib
import json


# =========================================================
# REQUIRED STATE
# =========================================================

required_cell22_names = [
    "project_root",
    "golden_v2_artifact_root",
    "provisional_seed_slot_records",
    "post_review_records_by_seed_id",
    "post_adjudication_accepted_seed_slot_ids",
    "final_rejected_seed_slot_ids",
    "source_repair_seed_slot_ids",
    "final_reject_adjudication_rows",
    "FULL_REVIEW_POLICY_VERSION",
    "PROMPT_VERSION",
    "FINAL_REJECT_ADJUDICATION_POLICY_VERSION",
]


missing_cell22_names = [
    name
    for name in required_cell22_names
    if name not in globals()
]


assert not missing_cell22_names, (
    "Missing required Post-Review objects: "
    + ", ".join(
        missing_cell22_names
    )
)


# =========================================================
# PRECONDITIONS
# =========================================================

assert (
    len(
        provisional_seed_slot_records
    )
    == 3817
)


assert (
    len(
        post_review_records_by_seed_id
    )
    == 3817
)


assert (
    len(
        post_adjudication_accepted_seed_slot_ids
    )
    == 3813
)


assert (
    len(
        final_rejected_seed_slot_ids
    )
    == 4
)


assert (
    len(
        source_repair_seed_slot_ids
    )
    == 4
)


assert (
    len(
        final_reject_adjudication_rows
    )
    == 4
)


assert (
    set(
        post_adjudication_accepted_seed_slot_ids
    ).isdisjoint(
        set(
            final_rejected_seed_slot_ids
        )
    )
)


assert (
    (
        set(
            post_adjudication_accepted_seed_slot_ids
        )
        | set(
            final_rejected_seed_slot_ids
        )
    )
    == set(
        post_review_records_by_seed_id
    )
)


# =========================================================
# PATHS
# =========================================================

post_review_judgments_path = (
    golden_v2_artifact_root
    / "golden_v2_seed_suitability_judgments_v1.jsonl"
)


reviewed_seed_pool_path = (
    golden_v2_artifact_root
    / "golden_v2_seed_pool_reviewed_v1.jsonl"
)


post_review_report_path = (
    golden_v2_artifact_root
    / "golden_v2_seed_suitability_report_v1.json"
)


# =========================================================
# HELPERS
# =========================================================

def canonical_json_line(
    obj,
):

    return (
        json.dumps(
            obj,
            ensure_ascii=False,
            sort_keys=True,
            separators=(",", ":"),
        )
        + "\n"
    )


def canonical_pretty_json(
    obj,
):

    return (
        json.dumps(
            obj,
            ensure_ascii=False,
            indent=2,
            sort_keys=True,
        )
        + "\n"
    )


def sha256_text(
    text,
):

    return (
        hashlib.sha256(
            text.encode(
                "utf-8"
            )
        ).hexdigest()
    )


def sha256_path(
    path,
):

    digest = (
        hashlib.sha256()
    )


    with path.open(
        "rb"
    ) as handle:

        for chunk in iter(
            lambda: handle.read(
                1024 * 1024
            ),
            b"",
        ):

            digest.update(
                chunk
            )


    return (
        digest.hexdigest()
    )


def freeze_exact_text(
    path,
    text,
):

    encoded = (
        text.encode(
            "utf-8"
        )
    )


    if path.exists():

        existing = (
            path.read_bytes()
        )


        assert (
            existing
            == encoded
        ), (
            "Frozen artifact already exists "
            "with different content:\n"
            f"{path}"
        )


        return (
            "REUSED_IDENTICAL_FROZEN"
        )


    path.parent.mkdir(
        parents=True,
        exist_ok=True,
    )


    temporary_path = (
        path.with_name(
            path.name
            + ".tmp"
        )
    )


    temporary_path.write_bytes(
        encoded
    )


    temporary_path.replace(
        path
    )


    return (
        "CREATED_FROZEN"
    )


# =========================================================
# INDEX PROVISIONAL SEED RECORDS
# =========================================================

provisional_seed_by_id = {}


for record in (
    provisional_seed_slot_records
):

    seed_slot_id = (
        record.get(
            "seed_slot_id"
        )
    )


    assert isinstance(
        seed_slot_id,
        str,
    )


    assert (
        seed_slot_id
        not in provisional_seed_by_id
    )


    provisional_seed_by_id[
        seed_slot_id
    ] = (
        record
    )


assert (
    len(
        provisional_seed_by_id
    )
    == 3817
)


assert (
    set(
        provisional_seed_by_id
    )
    == set(
        post_review_records_by_seed_id
    )
)


# =========================================================
# INDEX REJECT ADJUDICATION
# =========================================================

reject_adjudication_by_seed_id = {
    row[
        "seed_slot_id"
    ]: row
    for row in (
        final_reject_adjudication_rows
    )
}


assert (
    set(
        reject_adjudication_by_seed_id
    )
    == set(
        final_rejected_seed_slot_ids
    )
)


# =========================================================
# BUILD 3817 FINAL JUDGMENTS
# =========================================================

golden_v2_seed_suitability_judgments = []


accepted_set = set(
    post_adjudication_accepted_seed_slot_ids
)


rejected_set = set(
    final_rejected_seed_slot_ids
)


repair_set = set(
    source_repair_seed_slot_ids
)


for seed_slot_id in sorted(
    provisional_seed_by_id
):

    review_record = (
        post_review_records_by_seed_id[
            seed_slot_id
        ]
    )


    parsed_json = (
        review_record[
            "parsed_json"
        ]
    )


    semantic_verdict = (
        parsed_json[
            "verdict"
        ]
    )


    if seed_slot_id in accepted_set:

        final_seed_decision = (
            "ACCEPT_FOR_GOLDEN_V2_SEED_POOL"
        )

        repair_required = (
            False
        )

        repair_mode = (
            None
        )

        adjudication_reason = (
            None
        )

        source_knowledge_decision = (
            "NOT_APPLICABLE"
        )

        adjudication_policy_version = (
            None
        )


    elif seed_slot_id in rejected_set:

        reject_adjudication = (
            reject_adjudication_by_seed_id[
                seed_slot_id
            ]
        )


        final_seed_decision = (
            "REJECT_CURRENT_SEED"
        )

        repair_required = (
            seed_slot_id
            in repair_set
        )

        repair_mode = (
            reject_adjudication[
                "repair_mode"
            ]
        )

        adjudication_reason = (
            reject_adjudication[
                "adjudication_reason"
            ]
        )

        source_knowledge_decision = (
            reject_adjudication[
                "source_knowledge_decision"
            ]
        )

        adjudication_policy_version = (
            reject_adjudication[
                "policy_version"
            ]
        )


    else:

        raise AssertionError(
            "Unaccounted Seed Slot: "
            + seed_slot_id
        )


    judgment = {
        "seed_slot_id": (
            seed_slot_id
        ),

        "semantic_review": {
            "policy_version": (
                FULL_REVIEW_POLICY_VERSION
            ),

            "prompt_version": (
                PROMPT_VERSION
            ),

            "verdict": (
                semantic_verdict
            ),

            "dimensions": (
                parsed_json[
                    "dimensions"
                ]
            ),

            "brief_reason_fa": (
                parsed_json[
                    "brief_reason_fa"
                ]
            ),

            "canonical_reason_codes": (
                review_record.get(
                    "canonical_reason_codes"
                )
                or []
            ),
        },

        "post_review_decision": {
            "final_seed_decision": (
                final_seed_decision
            ),

            "repair_required": (
                repair_required
            ),

            "repair_mode": (
                repair_mode
            ),

            "source_knowledge_decision": (
                source_knowledge_decision
            ),

            "adjudication_reason": (
                adjudication_reason
            ),

            "adjudication_policy_version": (
                adjudication_policy_version
            ),
        },
    }


    golden_v2_seed_suitability_judgments.append(
        judgment
    )


# =========================================================
# JUDGMENT VALIDATION
# =========================================================

assert (
    len(
        golden_v2_seed_suitability_judgments
    )
    == 3817
)


assert (
    len({
        row[
            "seed_slot_id"
        ]
        for row in (
            golden_v2_seed_suitability_judgments
        )
    })
    == 3817
)


final_decision_counts = Counter(
    row[
        "post_review_decision"
    ][
        "final_seed_decision"
    ]
    for row in (
        golden_v2_seed_suitability_judgments
    )
)


assert (
    final_decision_counts[
        "ACCEPT_FOR_GOLDEN_V2_SEED_POOL"
    ]
    == 3813
)


assert (
    final_decision_counts[
        "REJECT_CURRENT_SEED"
    ]
    == 4
)


repair_required_count = sum(
    row[
        "post_review_decision"
    ][
        "repair_required"
    ]
    is True
    for row in (
        golden_v2_seed_suitability_judgments
    )
)


assert (
    repair_required_count
    == 4
)


# =========================================================
# BUILD REVIEWED 3813-SEED POOL
#
# Preserve original Seed records exactly and append one
# deterministic decision block.
# =========================================================

golden_v2_seed_pool_reviewed = []


for seed_slot_id in sorted(
    post_adjudication_accepted_seed_slot_ids
):

    source_record = (
        provisional_seed_by_id[
            seed_slot_id
        ]
    )


    review_record = (
        post_review_records_by_seed_id[
            seed_slot_id
        ]
    )


    parsed_json = (
        review_record[
            "parsed_json"
        ]
    )


    reviewed_record = dict(
        source_record
    )


    assert (
        "post_review_decision"
        not in reviewed_record
    ), (
        "Unexpected pre-existing post_review_decision "
        f"in Seed {seed_slot_id}"
    )


    reviewed_record[
        "post_review_decision"
    ] = {
        "final_seed_decision": (
            "ACCEPT_FOR_GOLDEN_V2_SEED_POOL"
        ),

        "semantic_verdict": (
            parsed_json[
                "verdict"
            ]
        ),

        "semantic_review_policy_version": (
            FULL_REVIEW_POLICY_VERSION
        ),

        "semantic_prompt_version": (
            PROMPT_VERSION
        ),
    }


    golden_v2_seed_pool_reviewed.append(
        reviewed_record
    )


assert (
    len(
        golden_v2_seed_pool_reviewed
    )
    == 3813
)


assert (
    len({
        row[
            "seed_slot_id"
        ]
        for row in (
            golden_v2_seed_pool_reviewed
        )
    })
    == 3813
)


assert (
    {
        row[
            "seed_slot_id"
        ]
        for row in (
            golden_v2_seed_pool_reviewed
        )
    }
    == accepted_set
)


# =========================================================
# SERIALIZE JUDGMENTS + REVIEWED POOL
# =========================================================

judgments_text = "".join(
    canonical_json_line(
        row
    )
    for row in (
        golden_v2_seed_suitability_judgments
    )
)


reviewed_seed_pool_text = "".join(
    canonical_json_line(
        row
    )
    for row in (
        golden_v2_seed_pool_reviewed
    )
)


judgments_sha256 = (
    sha256_text(
        judgments_text
    )
)


reviewed_seed_pool_sha256 = (
    sha256_text(
        reviewed_seed_pool_text
    )
)


# =========================================================
# BUILD SUMMARY REPORT
# =========================================================

golden_v2_seed_suitability_report = {
    "report_version": (
        "golden_v2_seed_suitability_report_v1"
    ),

    "stage": (
        "POST_REVIEW_SEED_ADJUDICATION"
    ),

    "status": (
        "PASS"
    ),

    "population": {
        "provisional_seed_slots": (
            3817
        ),

        "semantic_reviews_valid": (
            3817
        ),

        "accepted_current_seeds": (
            3813
        ),

        "rejected_current_seeds": (
            4
        ),

        "source_repair_candidates": (
            4
        ),

        "population_accounted": (
            3817
        ),
    },

    "semantic_review": {
        "policy_version": (
            FULL_REVIEW_POLICY_VERSION
        ),

        "prompt_version": (
            PROMPT_VERSION
        ),

        "verdict_counts": dict(
            sorted(
                final_verdict_counts.items()
            )
        ),
    },

    "reject_adjudication": {
        "policy_version": (
            FINAL_REJECT_ADJUDICATION_POLICY_VERSION
        ),

        "rejected_current_seed_slot_ids": (
            sorted(
                final_rejected_seed_slot_ids
            )
        ),

        "source_repair_seed_slot_ids": (
            sorted(
                source_repair_seed_slot_ids
            )
        ),

        "all_rejected_source_knowledge_preserved": (
            True
        ),

        "frozen_evidence_silently_expanded": (
            False
        ),
    },

    "artifacts": {
        "judgments": {
            "path": str(
                post_review_judgments_path.relative_to(
                    project_root
                )
            ),

            "records": (
                3817
            ),

            "sha256": (
                judgments_sha256
            ),
        },

        "reviewed_seed_pool": {
            "path": str(
                reviewed_seed_pool_path.relative_to(
                    project_root
                )
            ),

            "records": (
                3813
            ),

            "sha256": (
                reviewed_seed_pool_sha256
            ),
        },
    },

    "immutability": {
        "semantic_review_checkpoints_modified": (
            False
        ),

        "provisional_seed_records_modified_in_place": (
            False
        ),

        "rejected_seed_evidence_expanded_in_place": (
            False
        ),
    },

    "next_stage": (
        "GOLDEN_V2_REPAIR_CANDIDATE_HANDLING_AND_"
        "QUESTION_GENERATION_READINESS"
    ),
}


post_review_report_text = (
    canonical_pretty_json(
        golden_v2_seed_suitability_report
    )
)


post_review_report_sha256 = (
    sha256_text(
        post_review_report_text
    )
)


# =========================================================
# FREEZE
# =========================================================

judgments_write_status = (
    freeze_exact_text(
        post_review_judgments_path,
        judgments_text,
    )
)


reviewed_seed_pool_write_status = (
    freeze_exact_text(
        reviewed_seed_pool_path,
        reviewed_seed_pool_text,
    )
)


post_review_report_write_status = (
    freeze_exact_text(
        post_review_report_path,
        post_review_report_text,
    )
)


# =========================================================
# INDEPENDENT DISK VALIDATION
# =========================================================

assert (
    sha256_path(
        post_review_judgments_path
    )
    == judgments_sha256
)


assert (
    sha256_path(
        reviewed_seed_pool_path
    )
    == reviewed_seed_pool_sha256
)


assert (
    sha256_path(
        post_review_report_path
    )
    == post_review_report_sha256
)


# =========================================================
# READ BACK JSONL
# =========================================================

disk_judgments = []


with post_review_judgments_path.open(
    "r",
    encoding="utf-8",
) as handle:

    for line in handle:

        if not line.strip():
            continue

        disk_judgments.append(
            json.loads(
                line
            )
        )


disk_reviewed_seed_pool = []


with reviewed_seed_pool_path.open(
    "r",
    encoding="utf-8",
) as handle:

    for line in handle:

        if not line.strip():
            continue

        disk_reviewed_seed_pool.append(
            json.loads(
                line
            )
        )


disk_post_review_report = (
    json.loads(
        post_review_report_path.read_text(
            encoding="utf-8"
        )
    )
)


# =========================================================
# FINAL ARTIFACT CHECKS
# =========================================================

post_review_artifact_checks = {
    "judgments_disk_records_3817": (
        len(
            disk_judgments
        )
        == 3817
    ),

    "reviewed_seed_pool_disk_records_3813": (
        len(
            disk_reviewed_seed_pool
        )
        == 3813
    ),

    "judgment_seed_ids_unique": (
        len({
            row[
                "seed_slot_id"
            ]
            for row in (
                disk_judgments
            )
        })
        == 3817
    ),

    "reviewed_seed_ids_unique": (
        len({
            row[
                "seed_slot_id"
            ]
            for row in (
                disk_reviewed_seed_pool
            )
        })
        == 3813
    ),

    "accepted_population_exact": (
        {
            row[
                "seed_slot_id"
            ]
            for row in (
                disk_reviewed_seed_pool
            )
        }
        == accepted_set
    ),

    "rejected_not_in_reviewed_pool": (
        set(
            final_rejected_seed_slot_ids
        ).isdisjoint({
            row[
                "seed_slot_id"
            ]
            for row in (
                disk_reviewed_seed_pool
            )
        })
    ),

    "report_status_pass": (
        disk_post_review_report[
            "status"
        ]
        == "PASS"
    ),

    "report_population_accounted_3817": (
        disk_post_review_report[
            "population"
        ][
            "population_accounted"
        ]
        == 3817
    ),

    "report_accepted_3813": (
        disk_post_review_report[
            "population"
        ][
            "accepted_current_seeds"
        ]
        == 3813
    ),

    "report_rejected_4": (
        disk_post_review_report[
            "population"
        ][
            "rejected_current_seeds"
        ]
        == 4
    ),

    "report_repair_candidates_4": (
        disk_post_review_report[
            "population"
        ][
            "source_repair_candidates"
        ]
        == 4
    ),
}


post_review_artifact_status = (
    "PASS"
    if all(
        post_review_artifact_checks.values()
    )
    else "FAIL"
)


if (
    post_review_artifact_status
    == "PASS"
):

    golden_v2_post_review_artifact_readiness = (
        "READY_FOR_REPAIR_CANDIDATE_HANDLING"
    )

else:

    golden_v2_post_review_artifact_readiness = (
        "BLOCKED_POST_REVIEW_ARTIFACT_REPAIR_REQUIRED"
    )


# =========================================================
# OUTPUT
# =========================================================

print(
    "=== GOLDEN V2 POST-REVIEW DECISION ARTIFACTS ==="
)

print()

print(
    "Final judgment records:",
    len(
        disk_judgments
    ),
)

print(
    "Reviewed accepted Seed records:",
    len(
        disk_reviewed_seed_pool
    ),
)

print(
    "Rejected current Seeds:",
    len(
        final_rejected_seed_slot_ids
    ),
)

print(
    "Source repair candidates:",
    len(
        source_repair_seed_slot_ids
    ),
)


print()

print(
    "Judgments artifact:",
    post_review_judgments_path.relative_to(
        project_root
    ),
)

print(
    "Judgments SHA-256:",
    judgments_sha256,
)

print(
    "Judgments write status:",
    judgments_write_status,
)


print()

print(
    "Reviewed Seed Pool artifact:",
    reviewed_seed_pool_path.relative_to(
        project_root
    ),
)

print(
    "Reviewed Seed Pool SHA-256:",
    reviewed_seed_pool_sha256,
)

print(
    "Reviewed Seed Pool write status:",
    reviewed_seed_pool_write_status,
)


print()

print(
    "Report artifact:",
    post_review_report_path.relative_to(
        project_root
    ),
)

print(
    "Report SHA-256:",
    post_review_report_sha256,
)

print(
    "Report write status:",
    post_review_report_write_status,
)


print()

print(
    "Validation checks:"
)


for (
    check_name,
    check_value,
) in (
    post_review_artifact_checks.items()
):

    print(
        " -",
        check_name,
        ":",
        (
            "PASS"
            if check_value
            else "FAIL"
        ),
    )


print()

print(
    "API calls performed:",
    0,
)

print(
    "Semantic Review checkpoints modified:",
    0,
)

print(
    "Next readiness:",
    golden_v2_post_review_artifact_readiness,
)

print(
    "POST-REVIEW DECISION ARTIFACTS:",
    post_review_artifact_status,
)


assert (
    post_review_artifact_status
    == "PASS"
), (
    "Post-Review decision artifact validation failed."
)
````

Dataset خواهد بود.

````python
# Code Cell 23 — Final Golden Dataset Contract & Input Readiness
# =========================================================
# Goals:
# - Inspect the existing Golden v2 contract/schema artifacts
# - Inspect the frozen 3813-record Reviewed Seed Pool
# - Discover the actual input schema without guessing
# - Determine what is already available for final Golden records
# - Prepare the next construction step
#
# Safety:
# - ZERO API calls
# - ZERO file writes
# - ZERO artifact mutation
# - ZERO new Golden records
# =========================================================


from pathlib import Path
from collections import Counter
import json


# =========================================================
# REQUIRED STATE
# =========================================================

required_cell23_names = [
    "project_root",
    "golden_v2_artifact_root",
    "reviewed_seed_pool_path",
]


missing_cell23_names = [
    name
    for name in required_cell23_names
    if name not in globals()
]


assert not missing_cell23_names, (
    "Missing required Golden v2 objects: "
    + ", ".join(
        missing_cell23_names
    )
)


assert (
    golden_v2_artifact_root.exists()
), (
    "Golden v2 artifact root does not exist: "
    + str(
        golden_v2_artifact_root
    )
)


assert (
    reviewed_seed_pool_path.exists()
), (
    "Reviewed Seed Pool does not exist: "
    + str(
        reviewed_seed_pool_path
    )
)


# =========================================================
# LOAD FROZEN REVIEWED SEED POOL FROM DISK
#
# Do not trust only in-memory state.
# =========================================================

cell23_reviewed_seed_records = []


with reviewed_seed_pool_path.open(
    "r",
    encoding="utf-8",
) as handle:

    for line_number, line in enumerate(
        handle,
        start=1,
    ):

        if not line.strip():
            continue


        try:

            record = (
                json.loads(
                    line
                )
            )

        except Exception as exc:

            raise AssertionError(
                "Invalid JSON in Reviewed Seed Pool "
                f"at line {line_number}: {exc}"
            )


        assert isinstance(
            record,
            dict,
        )


        cell23_reviewed_seed_records.append(
            record
        )


assert (
    len(
        cell23_reviewed_seed_records
    )
    == 3813
), (
    "Reviewed Seed Pool population mismatch: "
    f"{len(cell23_reviewed_seed_records)}"
)


# =========================================================
# BASIC SEED POOL IDENTITY VALIDATION
# =========================================================

cell23_reviewed_seed_ids = [
    record.get(
        "seed_slot_id"
    )
    for record in (
        cell23_reviewed_seed_records
    )
]


assert all(
    isinstance(
        seed_slot_id,
        str,
    )
    and bool(
        seed_slot_id
    )
    for seed_slot_id in (
        cell23_reviewed_seed_ids
    )
)


assert (
    len(
        set(
            cell23_reviewed_seed_ids
        )
    )
    == 3813
), (
    "Reviewed Seed Pool contains duplicate Seed IDs."
)


# =========================================================
# DISCOVER ACTUAL SEED RECORD SCHEMA
# =========================================================

top_level_key_signatures = Counter(
    tuple(
        sorted(
            record.keys()
        )
    )
    for record in (
        cell23_reviewed_seed_records
    )
)


all_seed_top_level_keys = sorted({
    key
    for record in (
        cell23_reviewed_seed_records
    )
    for key in (
        record.keys()
    )
})


seed_field_presence_counts = {
    key: sum(
        key in record
        for record in (
            cell23_reviewed_seed_records
        )
    )
    for key in (
        all_seed_top_level_keys
    )
}


# =========================================================
# DISCOVER NESTED STRUCTURES
# =========================================================

nested_dict_field_keys = {}


for top_level_key in (
    all_seed_top_level_keys
):

    nested_keys = set()


    dict_record_count = 0


    for record in (
        cell23_reviewed_seed_records
    ):

        value = (
            record.get(
                top_level_key
            )
        )


        if isinstance(
            value,
            dict,
        ):

            dict_record_count += 1

            nested_keys.update(
                value.keys()
            )


    if dict_record_count:

        nested_dict_field_keys[
            top_level_key
        ] = {
            "dict_record_count": (
                dict_record_count
            ),

            "observed_keys": (
                sorted(
                    nested_keys
                )
            ),
        }


# =========================================================
# DISCOVER CONTRACT / SCHEMA / GUIDE FILES
#
# Read-only recursive discovery.
# Exclude runtime work/cache paths from contract candidates.
# =========================================================

contract_name_signals = [
    "contract",
    "schema",
    "guide",
    "spec",
    "policy",
    "definition",
    "golden",
]


excluded_path_signals = [
    "/work/",
    "/cache/",
    "/caches/",
    "__pycache__",
]


contract_candidate_paths = []


for path in sorted(
    golden_v2_artifact_root.rglob(
        "*"
    )
):

    if not path.is_file():
        continue


    relative_path = (
        path.relative_to(
            project_root
        )
    )


    relative_path_text = (
        str(
            relative_path
        )
        .replace(
            "\\",
            "/",
        )
    )


    lower_name = (
        path.name.lower()
    )


    lower_relative_path = (
        relative_path_text.lower()
    )


    if any(
        signal in lower_relative_path
        for signal in (
            excluded_path_signals
        )
    ):

        continue


    if any(
        signal in lower_name
        for signal in (
            contract_name_signals
        )
    ):

        contract_candidate_paths.append(
            path
        )


# =========================================================
# READ SMALL TEXT-LIKE CONTRACT CANDIDATES
#
# We only inspect text-like files and cap display size.
# No OCR / binary parsing / mutation.
# =========================================================

text_like_suffixes = {
    ".json",
    ".jsonl",
    ".md",
    ".txt",
    ".yaml",
    ".yml",
}


MAX_CONTRACT_PREVIEW_CHARS = 12000


contract_candidate_previews = []


for path in (
    contract_candidate_paths
):

    if (
        path.suffix.lower()
        not in text_like_suffixes
    ):

        continue


    try:

        text = (
            path.read_text(
                encoding="utf-8"
            )
        )

    except Exception as exc:

        contract_candidate_previews.append({
            "path": str(
                path.relative_to(
                    project_root
                )
            ),

            "read_status": (
                "UNREADABLE"
            ),

            "error": (
                str(
                    exc
                )
            ),

            "preview": (
                None
            ),
        })

        continue


    contract_candidate_previews.append({
        "path": str(
            path.relative_to(
                project_root
            )
        ),

        "read_status": (
            "READABLE"
        ),

        "size_chars": (
            len(
                text
            )
        ),

        "preview": (
            text[
                :MAX_CONTRACT_PREVIEW_CHARS
            ]
        ),
    })


# =========================================================
# SEARCH CONTRACT CONTENT FOR FINAL-DATASET SIGNALS
# =========================================================

final_dataset_signals = [
    "question",
    "answer",
    "expected_claim",
    "expected claims",
    "gold_document",
    "gold_documents",
    "gold_span",
    "gold_spans",
    "answerability",
    "abstention",
    "difficulty",
    "intent",
    "topic",
    "source_type",
    "reviewer",
    "final",
    "freeze",
    "dev",
    "test",
]


contract_signal_rows = []


for item in (
    contract_candidate_previews
):

    if (
        item[
            "read_status"
        ]
        != "READABLE"
    ):

        continue


    preview = (
        item[
            "preview"
        ]
    )


    lower_preview = (
        preview.lower()
    )


    matched_signals = [
        signal
        for signal in (
            final_dataset_signals
        )
        if signal in lower_preview
    ]


    contract_signal_rows.append({
        "path": (
            item[
                "path"
            ]
        ),

        "matched_signals": (
            matched_signals
        ),

        "matched_signal_count": (
            len(
                matched_signals
            )
        ),
    })


contract_signal_rows.sort(
    key=lambda row: (
        row[
            "matched_signal_count"
        ],
        row[
            "path"
        ],
    ),
    reverse=True,
)


# =========================================================
# IDENTIFY CURRENT POST-REVIEW ARTIFACTS
# =========================================================

expected_post_review_artifact_names = [
    "golden_v2_seed_suitability_judgments_v1.jsonl",
    "golden_v2_seed_pool_reviewed_v1.jsonl",
    "golden_v2_seed_suitability_report_v1.json",
]


post_review_artifact_presence = {}


for artifact_name in (
    expected_post_review_artifact_names
):

    artifact_path = (
        golden_v2_artifact_root
        / artifact_name
    )


    post_review_artifact_presence[
        artifact_name
    ] = (
        artifact_path.exists()
        and artifact_path.is_file()
    )


# =========================================================
# READINESS CHECKS
#
# This Cell intentionally does NOT claim that a final
# Golden Dataset contract has been found merely because
# candidate files exist.
# =========================================================

cell23_readiness_checks = {
    "reviewed_seed_pool_exists": (
        reviewed_seed_pool_path.exists()
    ),

    "reviewed_seed_population_3813": (
        len(
            cell23_reviewed_seed_records
        )
        == 3813
    ),

    "reviewed_seed_ids_unique": (
        len(
            set(
                cell23_reviewed_seed_ids
            )
        )
        == 3813
    ),

    "all_post_review_artifacts_present": (
        all(
            post_review_artifact_presence.values()
        )
    ),

    "at_least_one_contract_candidate_found": (
        len(
            contract_candidate_paths
        )
        > 0
    ),

    "at_least_one_contract_candidate_readable": (
        any(
            item[
                "read_status"
            ]
            == "READABLE"
            for item in (
                contract_candidate_previews
            )
        )
    ),
}


cell23_readiness_status = (
    "PASS"
    if all(
        cell23_readiness_checks.values()
    )
    else "FAIL"
)


if (
    cell23_readiness_status
    == "PASS"
):

    cell23_next_readiness = (
        "READY_FOR_EXACT_GOLDEN_CONTRACT_RESOLUTION"
    )

else:

    cell23_next_readiness = (
        "BLOCKED_GOLDEN_INPUT_OR_CONTRACT_DISCOVERY_REPAIR_REQUIRED"
    )


# =========================================================
# OUTPUT
# =========================================================

print(
    "=== FINAL GOLDEN DATASET CONTRACT & INPUT READINESS ==="
)

print()

print(
    "Reviewed Seed Pool:",
    reviewed_seed_pool_path.relative_to(
        project_root
    ),
)

print(
    "Reviewed Seed records:",
    len(
        cell23_reviewed_seed_records
    ),
)

print(
    "Unique Seed IDs:",
    len(
        set(
            cell23_reviewed_seed_ids
        )
    ),
)

print(
    "Observed top-level schema signatures:",
    len(
        top_level_key_signatures
    ),
)


print()

print(
    "TOP-LEVEL SEED FIELDS:"
)


for key in (
    all_seed_top_level_keys
):

    print(
        " -",
        key,
        ":",
        seed_field_presence_counts[
            key
        ],
        "/",
        len(
            cell23_reviewed_seed_records
        ),
    )


print()

print(
    "NESTED DICT STRUCTURES:"
)


if nested_dict_field_keys:

    for (
        field_name,
        field_info,
    ) in (
        nested_dict_field_keys.items()
    ):

        print(
            " -",
            field_name,
            "| dict records:",
            field_info[
                "dict_record_count"
            ],
            "| keys:",
            field_info[
                "observed_keys"
            ],
        )

else:

    print(
        " - none"
    )


print()

print(
    "POST-REVIEW ARTIFACT PRESENCE:"
)


for (
    artifact_name,
    exists,
) in (
    post_review_artifact_presence.items()
):

    print(
        " -",
        artifact_name,
        ":",
        (
            "PRESENT"
            if exists
            else "MISSING"
        ),
    )


print()

print(
    "Contract/schema candidate files:",
    len(
        contract_candidate_paths
    ),
)


for path in (
    contract_candidate_paths
):

    print(
        " -",
        path.relative_to(
            project_root
        ),
    )


print()

print(
    "CONTRACT SIGNAL RANKING:"
)


if contract_signal_rows:

    for row in (
        contract_signal_rows[
            :20
        ]
    ):

        print(
            " -",
            row[
                "matched_signal_count"
            ],
            "|",
            row[
                "path"
            ],
            "|",
            row[
                "matched_signals"
            ],
        )

else:

    print(
        " - no readable contract signals found"
    )


print()

print(
    "READABLE CONTRACT PREVIEWS:"
)


for item in (
    contract_candidate_previews
):

    print(
        "=" * 70
    )

    print()

    print(
        "PATH:",
        item[
            "path"
        ],
    )

    print(
        "STATUS:",
        item[
            "read_status"
        ],
    )


    if (
        item[
            "read_status"
        ]
        == "READABLE"
    ):

        print(
            "SIZE CHARS:",
            item[
                "size_chars"
            ],
        )

        print()

        print(
            item[
                "preview"
            ]
        )

    else:

        print(
            "ERROR:",
            item[
                "error"
            ],
        )


    print()


print(
    "Validation checks:"
)


for (
    check_name,
    check_value,
) in (
    cell23_readiness_checks.items()
):

    print(
        " -",
        check_name,
        ":",
        (
            "PASS"
            if check_value
            else "FAIL"
        ),
    )


print()

print(
    "API calls performed:",
    0,
)

print(
    "Files modified:",
    0,
)

print(
    "Golden records created:",
    0,
)

print(
    "Reviewed Seed population:",
    len(
        cell23_reviewed_seed_records
    ),
)

print(
    "Next readiness:",
    cell23_next_readiness,
)

print(
    "FINAL GOLDEN DATASET CONTRACT & INPUT READINESS:",
    cell23_readiness_status,
)


assert (
    cell23_readiness_status
    == "PASS"
), (
    "Golden Dataset contract/input readiness failed."
)
````

````python
# Code Cell 24 — Evidence Resolution & Golden Record Schema Discovery
# =========================================================
# Goals:
# - Resolve exact Evidence records behind all 3813 accepted Seeds
# - Follow the real Registry lineage:
#       Seed -> Canonical Claim -> raw_claim_ids -> Evidence Links
# - Verify Evidence-ID agreement with the Reviewed Seed Pool
# - Inspect historical Golden v1 / Core-46 schemas
# - Prepare the evidence-grounded Golden v2 record contract
#
# Safety:
# - ZERO API calls
# - ZERO file writes
# - ZERO artifact mutation
# - ZERO question generation
# =========================================================

from pathlib import Path
from collections import Counter, defaultdict
import json


# =========================================================
# REQUIRED STATE FROM CELL 23
# =========================================================

required_cell24_names = [
    "project_root",
    "golden_v2_artifact_root",
    "cell23_reviewed_seed_records",
]

missing_cell24_names = [
    name
    for name in required_cell24_names
    if name not in globals()
]

assert not missing_cell24_names, (
    "Missing required Cell 23 state: "
    + ", ".join(missing_cell24_names)
)

assert len(cell23_reviewed_seed_records) == 3813


# =========================================================
# LOAD FROZEN GOLDEN V2 CONTRACT
# =========================================================

cell24_contract_path = (
    golden_v2_artifact_root
    / "golden_dataset_v2_registry_driven_contract_v1.json"
)

assert cell24_contract_path.exists(), (
    f"Golden v2 contract missing: {cell24_contract_path}"
)

with cell24_contract_path.open(
    "r",
    encoding="utf-8",
) as handle:
    cell24_contract = json.load(handle)

assert isinstance(cell24_contract, dict)

source_manifests = cell24_contract.get(
    "source_manifests",
    {},
)

claim_registry_manifest = source_manifests.get(
    "claim_registry",
    {},
)

legacy_manifest = source_manifests.get(
    "legacy_golden_and_review",
    {},
)


# =========================================================
# RESOLVE PATHS ONLY FROM FROZEN CONTRACT
# =========================================================

required_registry_artifact_keys = [
    "canonical_claims",
    "evidence_links",
]

missing_registry_manifest_keys = [
    key
    for key in required_registry_artifact_keys
    if key not in claim_registry_manifest
]

assert not missing_registry_manifest_keys, (
    "Frozen contract is missing Registry artifact definitions: "
    + ", ".join(missing_registry_manifest_keys)
)

canonical_claims_path = (
    project_root
    / claim_registry_manifest[
        "canonical_claims"
    ][
        "path"
    ]
)

evidence_links_path = (
    project_root
    / claim_registry_manifest[
        "evidence_links"
    ][
        "path"
    ]
)

assert canonical_claims_path.exists(), (
    f"Canonical Claims artifact missing: {canonical_claims_path}"
)

assert evidence_links_path.exists(), (
    f"Evidence Links artifact missing: {evidence_links_path}"
)


# =========================================================
# GENERIC JSONL LOADER
# =========================================================

def cell24_load_jsonl(path):
    records = []

    with path.open(
        "r",
        encoding="utf-8",
    ) as handle:

        for line_number, line in enumerate(
            handle,
            start=1,
        ):

            if not line.strip():
                continue

            try:
                record = json.loads(line)

            except Exception as exc:
                raise AssertionError(
                    f"Invalid JSONL at {path}, "
                    f"line {line_number}: {exc}"
                )

            assert isinstance(record, dict), (
                f"Non-object JSONL record at "
                f"{path}, line {line_number}"
            )

            records.append(record)

    return records


# =========================================================
# LOAD REGISTRY CLAIMS + EVIDENCE LINKS
# =========================================================

cell24_canonical_claims = cell24_load_jsonl(
    canonical_claims_path
)

cell24_evidence_links = cell24_load_jsonl(
    evidence_links_path
)

assert len(cell24_canonical_claims) == 4006, (
    "Canonical Claim population mismatch: "
    f"{len(cell24_canonical_claims)}"
)

assert len(cell24_evidence_links) > 0


# =========================================================
# DISCOVER ACTUAL REGISTRY SCHEMAS
# =========================================================

canonical_claim_schema_signatures = Counter(
    tuple(sorted(record.keys()))
    for record in cell24_canonical_claims
)

evidence_link_schema_signatures = Counter(
    tuple(sorted(record.keys()))
    for record in cell24_evidence_links
)

canonical_claim_all_keys = sorted({
    key
    for record in cell24_canonical_claims
    for key in record.keys()
})

evidence_link_all_keys = sorted({
    key
    for record in cell24_evidence_links
    for key in record.keys()
})


# =========================================================
# REQUIRED REAL LINEAGE FIELDS
#
# Actual observed Registry path:
#
# Seed
#   -> representative_claim_id
#   -> Canonical Claim
#   -> raw_claim_ids
#   -> Evidence Link.raw_claim_id
#   -> Evidence Link.evidence_id
# =========================================================

assert all(
    "claim_id" in record
    for record in cell24_canonical_claims
), (
    "Canonical Claims do not consistently contain claim_id."
)

assert all(
    "raw_claim_ids" in record
    for record in cell24_canonical_claims
), (
    "Canonical Claims do not consistently contain raw_claim_ids."
)

assert all(
    "raw_claim_id" in record
    for record in cell24_evidence_links
), (
    "Evidence Links do not consistently contain raw_claim_id."
)

assert all(
    "evidence_id" in record
    for record in cell24_evidence_links
), (
    "Evidence Links do not consistently contain evidence_id."
)


# =========================================================
# INDEX CANONICAL CLAIMS
# =========================================================

canonical_claim_by_id = {}

canonical_claim_duplicate_ids = []

for record in cell24_canonical_claims:

    claim_id = record.get(
        "claim_id"
    )

    assert isinstance(
        claim_id,
        str,
    ) and claim_id

    if claim_id in canonical_claim_by_id:
        canonical_claim_duplicate_ids.append(
            claim_id
        )

    canonical_claim_by_id[
        claim_id
    ] = record

assert not canonical_claim_duplicate_ids


# =========================================================
# INDEX EVIDENCE LINKS BY EVIDENCE ID + RAW CLAIM ID
# =========================================================

evidence_by_id = {}

evidence_by_raw_claim_id = defaultdict(list)

duplicate_evidence_ids = []

for record in cell24_evidence_links:

    raw_claim_id = record.get(
        "raw_claim_id"
    )

    evidence_id = record.get(
        "evidence_id"
    )

    assert isinstance(
        raw_claim_id,
        str,
    ) and raw_claim_id

    assert isinstance(
        evidence_id,
        str,
    ) and evidence_id

    if evidence_id in evidence_by_id:
        duplicate_evidence_ids.append(
            evidence_id
        )

    evidence_by_id[
        evidence_id
    ] = record

    evidence_by_raw_claim_id[
        raw_claim_id
    ].append(
        record
    )

assert not duplicate_evidence_ids


# =========================================================
# RESOLVE ALL ACCEPTED SEEDS
#
# Resolution is validated from BOTH directions:
#
# 1. Seed representative_claim.evidence_ids
# 2. Canonical Claim raw_claim_ids -> Evidence Links
#
# Both sets must agree exactly.
# =========================================================

cell24_resolved_seed_inputs = []

cell24_resolution_errors = []

cell24_evidence_set_mismatches = []

for seed in cell23_reviewed_seed_records:

    seed_slot_id = seed.get(
        "seed_slot_id"
    )

    representative_claim_id = seed.get(
        "representative_claim_id"
    )

    representative_claim = seed.get(
        "representative_claim",
        {},
    )

    if not isinstance(
        representative_claim,
        dict,
    ):

        cell24_resolution_errors.append({
            "seed_slot_id": seed_slot_id,
            "error": "REPRESENTATIVE_CLAIM_NOT_DICT",
        })

        continue

    expected_evidence_ids = list(
        representative_claim.get(
            "evidence_ids",
            [],
        )
    )

    canonical_claim = canonical_claim_by_id.get(
        representative_claim_id
    )

    if canonical_claim is None:

        cell24_resolution_errors.append({
            "seed_slot_id": seed_slot_id,
            "error": "REPRESENTATIVE_CLAIM_NOT_FOUND_IN_REGISTRY",
            "representative_claim_id": representative_claim_id,
        })

        continue

    canonical_raw_claim_ids = canonical_claim.get(
        "raw_claim_ids",
        [],
    )

    if not isinstance(
        canonical_raw_claim_ids,
        list,
    ) or not canonical_raw_claim_ids:

        cell24_resolution_errors.append({
            "seed_slot_id": seed_slot_id,
            "error": "CANONICAL_CLAIM_HAS_NO_RAW_CLAIM_IDS",
            "representative_claim_id": representative_claim_id,
        })

        continue

    lineage_evidence_records = []

    for raw_claim_id in canonical_raw_claim_ids:

        lineage_evidence_records.extend(
            evidence_by_raw_claim_id.get(
                raw_claim_id,
                [],
            )
        )

    if not lineage_evidence_records:

        cell24_resolution_errors.append({
            "seed_slot_id": seed_slot_id,
            "error": "NO_EVIDENCE_FOUND_THROUGH_RAW_CLAIM_LINEAGE",
            "representative_claim_id": representative_claim_id,
            "raw_claim_ids": canonical_raw_claim_ids,
        })

        continue

    lineage_evidence_by_id = {
        record[
            "evidence_id"
        ]: record
        for record in lineage_evidence_records
    }

    lineage_evidence_ids = sorted(
        lineage_evidence_by_id.keys()
    )

    expected_evidence_ids_sorted = sorted(
        expected_evidence_ids
    )

    missing_expected_evidence_ids = [
        evidence_id
        for evidence_id in expected_evidence_ids_sorted
        if evidence_id not in evidence_by_id
    ]

    if missing_expected_evidence_ids:

        cell24_resolution_errors.append({
            "seed_slot_id": seed_slot_id,
            "error": "EXPECTED_EVIDENCE_ID_NOT_FOUND_GLOBALLY",
            "missing_evidence_ids": missing_expected_evidence_ids,
        })

        continue

    if (
        set(expected_evidence_ids_sorted)
        != set(lineage_evidence_ids)
    ):

        cell24_evidence_set_mismatches.append({
            "seed_slot_id": seed_slot_id,
            "representative_claim_id": representative_claim_id,
            "expected_evidence_ids": expected_evidence_ids_sorted,
            "lineage_evidence_ids": lineage_evidence_ids,
            "raw_claim_ids": canonical_raw_claim_ids,
        })

        continue

    resolved_evidence = [
        lineage_evidence_by_id[
            evidence_id
        ]
        for evidence_id in expected_evidence_ids_sorted
    ]

    cell24_resolved_seed_inputs.append({
        "seed_slot_id": seed_slot_id,

        "representative_claim_id": (
            representative_claim_id
        ),

        "canonical_claim": (
            canonical_claim
        ),

        "raw_claim_ids": list(
            canonical_raw_claim_ids
        ),

        "expected_evidence_ids": (
            expected_evidence_ids_sorted
        ),

        "resolved_evidence": (
            resolved_evidence
        ),

        "source_ids": (
            representative_claim.get(
                "source_ids",
                [],
            )
        ),

        "source_document_ids": (
            representative_claim.get(
                "source_document_ids",
                [],
            )
        ),

        "source_types": (
            representative_claim.get(
                "source_types",
                [],
            )
        ),

        "authority_class": (
            representative_claim.get(
                "authority_class"
            )
        ),

        "claim_types": (
            representative_claim.get(
                "claim_types",
                [],
            )
        ),
    })


# =========================================================
# HISTORICAL GOLDEN SCHEMA DISCOVERY
#
# Historical only:
# - Golden v1 = baseline
# - Core-46 = quality/regression anchor
#
# They do NOT automatically define Golden v2.
# =========================================================

historical_golden_sources = {}

for key in [
    "golden_v1",
    "core46",
]:

    item = legacy_manifest.get(
        key
    )

    if not isinstance(
        item,
        dict,
    ):
        continue

    relative_path = item.get(
        "path"
    )

    if not isinstance(
        relative_path,
        str,
    ):
        continue

    path = (
        project_root
        / relative_path
    )

    historical_golden_sources[
        key
    ] = {
        "path": path,
        "manifest": item,
    }


historical_schema_summary = {}

for (
    source_name,
    source_info,
) in historical_golden_sources.items():

    path = source_info[
        "path"
    ]

    if not path.exists():

        historical_schema_summary[
            source_name
        ] = {
            "status": "MISSING",
            "path": str(
                path.relative_to(
                    project_root
                )
            ),
        }

        continue

    records = cell24_load_jsonl(
        path
    )

    signatures = Counter(
        tuple(
            sorted(
                record.keys()
            )
        )
        for record in records
    )

    all_keys = sorted({
        key
        for record in records
        for key in record.keys()
    })

    field_presence = {
        key: sum(
            key in record
            for record in records
        )
        for key in all_keys
    }

    historical_schema_summary[
        source_name
    ] = {
        "status": "READABLE",

        "path": str(
            path.relative_to(
                project_root
            )
        ),

        "records": len(
            records
        ),

        "schema_signature_count": len(
            signatures
        ),

        "schema_signatures": signatures,

        "all_keys": all_keys,

        "field_presence": field_presence,

        "sample_records": records[
            :3
        ],
    }


# =========================================================
# RESOLVED EVIDENCE FIELD COVERAGE
# =========================================================

resolved_evidence_records = [
    evidence
    for item in cell24_resolved_seed_inputs
    for evidence in item[
        "resolved_evidence"
    ]
]

resolved_evidence_all_keys = sorted({
    key
    for record in resolved_evidence_records
    for key in record.keys()
})

resolved_evidence_field_presence = {
    key: sum(
        key in record
        for record in resolved_evidence_records
    )
    for key in resolved_evidence_all_keys
}


# =========================================================
# SOURCE-LINEAGE CONSISTENCY DIAGNOSTICS
# =========================================================

source_id_match_count = 0

document_id_match_count = 0

for item in cell24_resolved_seed_inputs:

    expected_source_ids = set(
        item[
            "source_ids"
        ]
    )

    expected_document_ids = set(
        item[
            "source_document_ids"
        ]
    )

    evidence_source_ids = {
        evidence.get(
            "source_id"
        )
        for evidence in item[
            "resolved_evidence"
        ]
        if evidence.get(
            "source_id"
        )
    }

    evidence_document_ids = {
        evidence.get(
            "document_id"
        )
        for evidence in item[
            "resolved_evidence"
        ]
        if evidence.get(
            "document_id"
        )
    }

    if (
        expected_source_ids
        == evidence_source_ids
    ):
        source_id_match_count += 1

    if (
        expected_document_ids
        == evidence_document_ids
    ):
        document_id_match_count += 1


# =========================================================
# CONTRACT REQUIREMENTS
# =========================================================

generation_policy = cell24_contract.get(
    "golden_generation_policy",
    {},
)

required_generation_contract_checks = {
    "answer_must_be_grounded": (
        generation_policy.get(
            "answer_must_be_grounded"
        )
        is True
    ),

    "evidence_first": (
        generation_policy.get(
            "evidence_first"
        )
        is True
    ),

    "preserve_claim_id": (
        generation_policy.get(
            "preserve_claim_id"
        )
        is True
    ),

    "preserve_evidence_id": (
        generation_policy.get(
            "preserve_evidence_id"
        )
        is True
    ),

    "preserve_source_lineage": (
        generation_policy.get(
            "preserve_source_lineage"
        )
        is True
    ),

    "question_supported_by_claim_and_evidence": (
        generation_policy.get(
            "question_must_be_supported_by_claim_and_evidence"
        )
        is True
    ),
}


# =========================================================
# VALIDATION CHECKS
# =========================================================

cell24_validation_checks = {
    "canonical_claim_population_4006": (
        len(
            cell24_canonical_claims
        )
        == 4006
    ),

    "canonical_claim_ids_unique": (
        len(
            canonical_claim_by_id
        )
        == 4006
    ),

    "evidence_links_loaded": (
        len(
            cell24_evidence_links
        )
        > 0
    ),

    "evidence_ids_unique": (
        len(
            evidence_by_id
        )
        == len(
            cell24_evidence_links
        )
    ),

    "real_lineage_uses_raw_claim_id": (
        "raw_claim_ids"
        in canonical_claim_all_keys
        and
        "raw_claim_id"
        in evidence_link_all_keys
    ),

    "accepted_seed_population_3813": (
        len(
            cell23_reviewed_seed_records
        )
        == 3813
    ),

    "all_3813_seed_inputs_resolved": (
        len(
            cell24_resolved_seed_inputs
        )
        == 3813
    ),

    "zero_seed_resolution_errors": (
        len(
            cell24_resolution_errors
        )
        == 0
    ),

    "zero_evidence_set_mismatches": (
        len(
            cell24_evidence_set_mismatches
        )
        == 0
    ),

    "resolved_evidence_nonempty": (
        len(
            resolved_evidence_records
        )
        > 0
    ),

    "all_seed_source_ids_match_evidence": (
        source_id_match_count
        == 3813
    ),

    "all_seed_document_ids_match_evidence": (
        document_id_match_count
        == 3813
    ),

    "contract_generation_requirements_pass": (
        all(
            required_generation_contract_checks.values()
        )
    ),

    "golden_v1_schema_discovered": (
        historical_schema_summary.get(
            "golden_v1",
            {},
        ).get(
            "status"
        )
        == "READABLE"
    ),

    "core46_schema_discovered": (
        historical_schema_summary.get(
            "core46",
            {},
        ).get(
            "status"
        )
        == "READABLE"
    ),
}


cell24_status = (
    "PASS"
    if all(
        cell24_validation_checks.values()
    )
    else "FAIL"
)

cell24_next_readiness = (
    "READY_FOR_GOLDEN_V2_RECORD_CONTRACT_DESIGN"
    if cell24_status == "PASS"
    else "BLOCKED_EVIDENCE_OR_SCHEMA_RESOLUTION_REPAIR_REQUIRED"
)


# =========================================================
# OUTPUT
# =========================================================

print(
    "=== GOLDEN V2 EVIDENCE RESOLUTION & SCHEMA DISCOVERY ==="
)

print()

print(
    "Canonical Claims:",
    len(
        cell24_canonical_claims
    ),
)

print(
    "Evidence Links:",
    len(
        cell24_evidence_links
    ),
)

print(
    "Accepted Seeds:",
    len(
        cell23_reviewed_seed_records
    ),
)

print(
    "Resolved Seed inputs:",
    len(
        cell24_resolved_seed_inputs
    ),
)

print(
    "Resolution errors:",
    len(
        cell24_resolution_errors
    ),
)

print(
    "Evidence-set mismatches:",
    len(
        cell24_evidence_set_mismatches
    ),
)

print(
    "Resolved Evidence records:",
    len(
        resolved_evidence_records
    ),
)

print()

print(
    "Resolved lineage:"
)

print(
    " Seed -> representative_claim_id"
)

print(
    " -> Canonical Claim.raw_claim_ids"
)

print(
    " -> Evidence Link.raw_claim_id"
)

print(
    " -> Evidence Link.evidence_id"
)

print()

print(
    "Source-ID exact matches:",
    source_id_match_count,
    "/ 3813",
)

print(
    "Document-ID exact matches:",
    document_id_match_count,
    "/ 3813",
)

print()

print(
    "CANONICAL CLAIM SCHEMA:"
)

print(
    "Schema signatures:",
    len(
        canonical_claim_schema_signatures
    ),
)

print(
    "Fields:",
    canonical_claim_all_keys,
)

print()

print(
    "EVIDENCE LINK SCHEMA:"
)

print(
    "Schema signatures:",
    len(
        evidence_link_schema_signatures
    ),
)

print(
    "Fields:",
    evidence_link_all_keys,
)

print()

print(
    "RESOLVED EVIDENCE FIELD COVERAGE:"
)

for key in resolved_evidence_all_keys:

    print(
        " -",
        key,
        ":",
        resolved_evidence_field_presence[
            key
        ],
        "/",
        len(
            resolved_evidence_records
        ),
    )

print()

print(
    "CONTRACT GENERATION REQUIREMENTS:"
)

for (
    check_name,
    check_value,
) in required_generation_contract_checks.items():

    print(
        " -",
        check_name,
        ":",
        (
            "PASS"
            if check_value
            else "FAIL"
        ),
    )

print()

print(
    "HISTORICAL GOLDEN SCHEMAS:"
)

for (
    source_name,
    summary,
) in historical_schema_summary.items():

    print(
        "=" * 70
    )

    print(
        "SOURCE:",
        source_name,
    )

    print(
        "STATUS:",
        summary[
            "status"
        ],
    )

    print(
        "PATH:",
        summary[
            "path"
        ],
    )

    if (
        summary[
            "status"
        ]
        != "READABLE"
    ):
        continue

    print(
        "RECORDS:",
        summary[
            "records"
        ],
    )

    print(
        "SCHEMA SIGNATURES:",
        summary[
            "schema_signature_count"
        ],
    )

    print(
        "FIELDS:",
        summary[
            "all_keys"
        ],
    )

    print(
        "FIELD PRESENCE:"
    )

    for (
        key,
        count,
    ) in summary[
        "field_presence"
    ].items():

        print(
            " -",
            key,
            ":",
            count,
            "/",
            summary[
                "records"
            ],
        )

    print()

    print(
        "SAMPLE RECORDS:"
    )

    for sample_index, sample in enumerate(
        summary[
            "sample_records"
        ],
        start=1,
    ):

        print()

        print(
            f"--- SAMPLE {sample_index} ---"
        )

        print(
            json.dumps(
                sample,
                ensure_ascii=False,
                indent=2,
            )
        )


print()

if cell24_resolution_errors:

    print(
        "RESOLUTION ERROR SAMPLE:"
    )

    for error in cell24_resolution_errors[
        :20
    ]:

        print(
            json.dumps(
                error,
                ensure_ascii=False,
            )
        )

    print()


if cell24_evidence_set_mismatches:

    print(
        "EVIDENCE-SET MISMATCH SAMPLE:"
    )

    for mismatch in cell24_evidence_set_mismatches[
        :20
    ]:

        print(
            json.dumps(
                mismatch,
                ensure_ascii=False,
            )
        )

    print()


print(
    "Validation checks:"
)

for (
    check_name,
    check_value,
) in cell24_validation_checks.items():

    print(
        " -",
        check_name,
        ":",
        (
            "PASS"
            if check_value
            else "FAIL"
        ),
    )

print()

print(
    "API calls performed:",
    0,
)

print(
    "Files modified:",
    0,
)

print(
    "Questions generated:",
    0,
)

print(
    "Golden records generated:",
    0,
)

print(
    "Next readiness:",
    cell24_next_readiness,
)

print(
    "GOLDEN V2 EVIDENCE RESOLUTION & SCHEMA DISCOVERY:",
    cell24_status,
)

assert cell24_status == "PASS", (
    "Golden v2 Evidence/Schema discovery failed."
)
````

<h2>مرحله ۲۵ — طراحی و تثبیت Contract رکورد Golden Dataset v2</h2>

<p>
مرحله قبل اتصال کامل ۳۸۱۳ Seed پذیرفته‌شده به Canonical Claimها و
۳۹۲۴ Evidence Record واقعی را اثبات کرد. تمام شناسه‌های Evidence،
Source و Document بدون mismatch از Registry بازیابی شدند.
</p>

<p>
اکنون پیش از تولید سؤال و پاسخ باید ساختار دقیق هر Golden Record مشخص شود.
این ساختار باید مستقیماً از نیازهای Contract اصلی Golden v2، ساختار واقعی
Claim Registry و Evidence Linkها، و تجربه نسخه‌های تاریخی Golden استخراج شود.
</p>

<p>
Golden v1 و Core-46 در این مرحله فقط Historical Baseline و Quality Anchor
هستند و Schema آن‌ها مستقیماً کپی نمی‌شود. Golden v2 باید شناسه‌های پایدار
Seed، Claim و Evidence، Source Lineage، سؤال، پاسخ مورد انتظار، نوع سؤال،
و Metadata لازم برای Validation و Benchmark آینده را به‌صورت صریح حفظ کند.
</p>

<p>
این مرحله فقط Contract پیشنهادی رکورد را در حافظه می‌سازد و آن را در برابر
داده‌های واقعی ۳۸۱۳ Seed اعتبارسنجی می‌کند. هنوز هیچ Question یا Answer
تولید نمی‌شود، هیچ API فراخوانی نمی‌شود و هیچ Artifactی روی دیسک نوشته
یا تغییر داده نمی‌شود.
</p>

</div>

````python
# Code Cell 25 — Golden v2 Record Contract Design
# =========================================================
# Goals:
# - Define the exact Golden v2 candidate-record contract
# - Derive it from:
#       frozen Golden v2 policy
#       real Claim Registry schema
#       real Evidence schema
#       historical Golden lessons
# - Validate that all deterministic provenance fields can
#   already be populated for all 3813 accepted Seeds
#
# Important:
# - Question/answer fields are declared but NOT generated here
# - No final split is assigned here
# - No retrieval chunk labels are invented here
#
# Safety:
# - ZERO API calls
# - ZERO file writes
# - ZERO artifact mutation
# - ZERO question generation
# =========================================================

from collections import Counter
import hashlib
import json


# =========================================================
# REQUIRED STATE FROM CELL 24
# =========================================================

required_cell25_names = [
    "project_root",
    "golden_v2_artifact_root",
    "cell24_contract",
    "cell24_resolved_seed_inputs",
    "cell23_reviewed_seed_records",
]

missing_cell25_names = [
    name
    for name in required_cell25_names
    if name not in globals()
]

assert not missing_cell25_names, (
    "Missing required Cell 24 state: "
    + ", ".join(missing_cell25_names)
)

assert len(cell24_resolved_seed_inputs) == 3813
assert len(cell23_reviewed_seed_records) == 3813


# =========================================================
# CONTRACT IDENTITY
# =========================================================

GOLDEN_V2_RECORD_CONTRACT_VERSION = (
    "golden_v2_record_contract_v1"
)

GOLDEN_V2_DATASET_CANDIDATE_VERSION = (
    "golden_dataset_v2_registry_driven_candidate_v1"
)

GOLDEN_V2_GENERATION_STATUS_INITIAL = (
    "PENDING_QUESTION_GENERATION"
)


# =========================================================
# FIELD GROUPS
#
# These groups separate:
# 1. stable identity
# 2. Registry provenance
# 3. exact evidence
# 4. generated evaluation content
# 5. generation metadata
# 6. validation / lifecycle metadata
# =========================================================

cell25_record_contract = {
    "contract_version": (
        GOLDEN_V2_RECORD_CONTRACT_VERSION
    ),

    "dataset_candidate_version": (
        GOLDEN_V2_DATASET_CANDIDATE_VERSION
    ),

    "record_granularity": (
        "ONE_GOLDEN_CANDIDATE_PER_ACCEPTED_SEED"
    ),

    "expected_candidate_population": 3813,

    "field_groups": {

        "identity": [
            "candidate_id",
            "dataset_version",
            "source_seed_slot_id",
            "claim_id",
        ],

        "claim": [
            "claim_text_fa",
            "claim_type",
            "claim_types",
            "authority_class",
        ],

        "evidence": [
            "evidence_ids",
            "evidence_quotes",
            "evidence_locators",
        ],

        "source_lineage": [
            "raw_claim_ids",
            "source_ids",
            "document_ids",
            "source_types",
        ],

        "evaluation_content": [
            "question",
            "expected_answer",
            "question_intent",
        ],

        "generation": [
            "generation",
            "generation_status",
        ],

        "validation": [
            "validation",
        ],
    },

    "required_fields": [
        "candidate_id",
        "dataset_version",
        "source_seed_slot_id",
        "claim_id",
        "claim_text_fa",
        "claim_type",
        "claim_types",
        "authority_class",
        "raw_claim_ids",
        "evidence_ids",
        "evidence_quotes",
        "evidence_locators",
        "source_ids",
        "document_ids",
        "source_types",
        "question",
        "expected_answer",
        "question_intent",
        "generation",
        "generation_status",
        "validation",
    ],

    "generation_required_fields": [
        "question",
        "expected_answer",
        "question_intent",
    ],

    "deterministic_pre_generation_fields": [
        "candidate_id",
        "dataset_version",
        "source_seed_slot_id",
        "claim_id",
        "claim_text_fa",
        "claim_type",
        "claim_types",
        "authority_class",
        "raw_claim_ids",
        "evidence_ids",
        "evidence_quotes",
        "evidence_locators",
        "source_ids",
        "document_ids",
        "source_types",
        "generation_status",
    ],

    "explicitly_deferred_fields": {
        "evaluation_split": (
            "DEFER_UNTIL_FINAL_DATASET_ACCEPTANCE_AND_SPLIT_POLICY"
        ),

        "retrieval_labels": (
            "DEFER_UNTIL_CHUNK_INDEX_MAPPING_IS_EXPLICITLY_VALIDATED"
        ),

        "taxonomy_topics": (
            "DO_NOT_COPY_FROM_LEGACY_DATASETS_WITHOUT_V2_MAPPING"
        ),

        "competency_ids": (
            "DO_NOT_COPY_FROM_LEGACY_DATASETS_WITHOUT_V2_MAPPING"
        ),

        "manual_decision": (
            "NOT_REQUIRED_FOR_MACHINE_CANDIDATE_CONSTRUCTION"
        ),

        "author_review": (
            "NOT_A_CURRENT_GATE"
        ),
    },

    "invariants": {
        "one_candidate_per_accepted_seed": True,

        "candidate_id_deterministic": True,

        "candidate_id_based_on_seed_and_contract": True,

        "claim_id_preserved": True,

        "evidence_ids_preserved": True,

        "evidence_quotes_exact": True,

        "source_lineage_preserved": True,

        "question_must_be_supported_by_claim_and_evidence": True,

        "answer_must_be_grounded": True,

        "no_external_knowledge_required": True,

        "question_must_be_standalone": True,

        "question_scope_must_be_unambiguous": True,

        "generation_does_not_mutate_registry": True,

        "generation_does_not_mutate_reviewed_seed_pool": True,
    },
}


# =========================================================
# CONTRACT CONSISTENCY
# =========================================================

all_declared_fields = []

for fields in (
    cell25_record_contract[
        "field_groups"
    ].values()
):
    all_declared_fields.extend(
        fields
    )

duplicate_declared_fields = sorted([
    field
    for field, count in Counter(
        all_declared_fields
    ).items()
    if count > 1
])

assert not duplicate_declared_fields, (
    "Duplicate fields across contract groups: "
    + ", ".join(
        duplicate_declared_fields
    )
)

assert set(
    cell25_record_contract[
        "required_fields"
    ]
) == set(
    all_declared_fields
), (
    "Required fields and grouped fields do not match."
)


# =========================================================
# DETERMINISTIC CANDIDATE ID
#
# Stable across reruns as long as:
# - Seed identity is unchanged
# - Contract version is unchanged
# =========================================================

def cell25_make_candidate_id(
    seed_slot_id,
):
    identity_payload = (
        GOLDEN_V2_RECORD_CONTRACT_VERSION
        + "|"
        + seed_slot_id
    )

    digest = hashlib.sha256(
        identity_payload.encode(
            "utf-8"
        )
    ).hexdigest()[:24]

    return (
        "golden_v2_candidate_"
        + digest
    )


# =========================================================
# NORMALIZATION HELPERS
# =========================================================

def cell25_unique_preserve_order(values):
    seen = set()
    result = []

    for value in values:

        if value in seen:
            continue

        seen.add(
            value
        )

        result.append(
            value
        )

    return result


def cell25_locator_copy(locator):
    if isinstance(
        locator,
        dict,
    ):
        return dict(
            locator
        )

    return locator


# =========================================================
# BUILD PRE-GENERATION RECORD SKELETONS
#
# Question/answer/intent remain None intentionally.
# These are NOT Golden Dataset records yet.
# =========================================================

cell25_candidate_skeletons = []

cell25_skeleton_errors = []

for item in cell24_resolved_seed_inputs:

    seed_slot_id = item[
        "seed_slot_id"
    ]

    canonical_claim = item[
        "canonical_claim"
    ]

    evidence_records = item[
        "resolved_evidence"
    ]

    claim_id = canonical_claim.get(
        "claim_id"
    )

    claim_text_fa = canonical_claim.get(
        "claim_text_fa"
    )

    raw_claim_ids = list(
        item[
            "raw_claim_ids"
        ]
    )

    evidence_ids = [
        evidence.get(
            "evidence_id"
        )
        for evidence in evidence_records
    ]

    evidence_quotes = [
        evidence.get(
            "evidence_quote_exact"
        )
        for evidence in evidence_records
    ]

    evidence_locators = [
        cell25_locator_copy(
            evidence.get(
                "locator"
            )
        )
        for evidence in evidence_records
    ]

    source_ids = cell25_unique_preserve_order([
        evidence.get(
            "source_id"
        )
        for evidence in evidence_records
        if evidence.get(
            "source_id"
        )
    ])

    document_ids = cell25_unique_preserve_order([
        evidence.get(
            "document_id"
        )
        for evidence in evidence_records
        if evidence.get(
            "document_id"
        )
    ])

    source_types = cell25_unique_preserve_order([
        evidence.get(
            "source_type"
        )
        for evidence in evidence_records
        if evidence.get(
            "source_type"
        )
    ])

    if not isinstance(
        claim_id,
        str,
    ) or not claim_id:

        cell25_skeleton_errors.append({
            "seed_slot_id": seed_slot_id,
            "error": "INVALID_CLAIM_ID",
        })

        continue

    if not isinstance(
        claim_text_fa,
        str,
    ) or not claim_text_fa.strip():

        cell25_skeleton_errors.append({
            "seed_slot_id": seed_slot_id,
            "error": "INVALID_CLAIM_TEXT",
        })

        continue

    if not evidence_ids:

        cell25_skeleton_errors.append({
            "seed_slot_id": seed_slot_id,
            "error": "NO_EVIDENCE_IDS",
        })

        continue

    if any(
        not isinstance(
            quote,
            str,
        )
        or not quote.strip()
        for quote in evidence_quotes
    ):

        cell25_skeleton_errors.append({
            "seed_slot_id": seed_slot_id,
            "error": "INVALID_EVIDENCE_QUOTE",
        })

        continue

    candidate_id = (
        cell25_make_candidate_id(
            seed_slot_id
        )
    )

    skeleton = {
        "candidate_id": (
            candidate_id
        ),

        "dataset_version": (
            GOLDEN_V2_DATASET_CANDIDATE_VERSION
        ),

        "source_seed_slot_id": (
            seed_slot_id
        ),

        "claim_id": (
            claim_id
        ),

        "claim_text_fa": (
            claim_text_fa
        ),

        "claim_type": (
            canonical_claim.get(
                "claim_type"
            )
        ),

        "claim_types": list(
            canonical_claim.get(
                "claim_types",
                [],
            )
        ),

        "authority_class": (
            canonical_claim.get(
                "authority_class"
            )
        ),

        "raw_claim_ids": (
            raw_claim_ids
        ),

        "evidence_ids": (
            evidence_ids
        ),

        "evidence_quotes": (
            evidence_quotes
        ),

        "evidence_locators": (
            evidence_locators
        ),

        "source_ids": (
            source_ids
        ),

        "document_ids": (
            document_ids
        ),

        "source_types": (
            source_types
        ),

        "question": None,

        "expected_answer": None,

        "question_intent": None,

        "generation": {
            "status": (
                "NOT_STARTED"
            ),

            "model": None,

            "prompt_version": None,

            "request_fingerprint": None,
        },

        "generation_status": (
            GOLDEN_V2_GENERATION_STATUS_INITIAL
        ),

        "validation": {
            "contract_version": (
                GOLDEN_V2_RECORD_CONTRACT_VERSION
            ),

            "pre_generation_structure_valid": True,

            "semantic_validation_status": (
                "NOT_STARTED"
            ),

            "final_acceptance_status": (
                "NOT_STARTED"
            ),
        },
    }

    cell25_candidate_skeletons.append(
        skeleton
    )


# =========================================================
# STRUCTURAL VALIDATION
# =========================================================

cell25_candidate_ids = [
    record[
        "candidate_id"
    ]
    for record in cell25_candidate_skeletons
]

cell25_seed_ids = [
    record[
        "source_seed_slot_id"
    ]
    for record in cell25_candidate_skeletons
]

cell25_claim_ids = [
    record[
        "claim_id"
    ]
    for record in cell25_candidate_skeletons
]


cell25_all_required_fields_present = all(
    set(
        cell25_record_contract[
            "required_fields"
        ]
    ).issubset(
        record.keys()
    )
    for record in cell25_candidate_skeletons
)


cell25_generation_fields_empty = all(
    record[
        "question"
    ] is None
    and record[
        "expected_answer"
    ] is None
    and record[
        "question_intent"
    ] is None
    and record[
        "generation_status"
    ]
    == GOLDEN_V2_GENERATION_STATUS_INITIAL
    for record in cell25_candidate_skeletons
)


# =========================================================
# VERIFY PROVENANCE AGAINST CELL 24 INPUTS
# =========================================================

cell24_input_by_seed_id = {
    item[
        "seed_slot_id"
    ]: item
    for item in cell24_resolved_seed_inputs
}

cell25_provenance_mismatches = []

for record in cell25_candidate_skeletons:

    seed_slot_id = record[
        "source_seed_slot_id"
    ]

    source_item = cell24_input_by_seed_id[
        seed_slot_id
    ]

    expected_evidence_ids = [
        evidence[
            "evidence_id"
        ]
        for evidence in source_item[
            "resolved_evidence"
        ]
    ]

    expected_source_ids = (
        cell25_unique_preserve_order([
            evidence.get(
                "source_id"
            )
            for evidence in source_item[
                "resolved_evidence"
            ]
            if evidence.get(
                "source_id"
            )
        ])
    )

    expected_document_ids = (
        cell25_unique_preserve_order([
            evidence.get(
                "document_id"
            )
            for evidence in source_item[
                "resolved_evidence"
            ]
            if evidence.get(
                "document_id"
            )
        ])
    )

    if (
        record[
            "evidence_ids"
        ]
        != expected_evidence_ids
        or
        record[
            "source_ids"
        ]
        != expected_source_ids
        or
        record[
            "document_ids"
        ]
        != expected_document_ids
    ):

        cell25_provenance_mismatches.append(
            seed_slot_id
        )


# =========================================================
# CONTRACT FINGERPRINT
# =========================================================

cell25_contract_canonical_json = (
    json.dumps(
        cell25_record_contract,
        ensure_ascii=False,
        sort_keys=True,
        separators=(
            ",",
            ":",
        ),
    )
)

cell25_record_contract_sha256 = (
    hashlib.sha256(
        cell25_contract_canonical_json.encode(
            "utf-8"
        )
    ).hexdigest()
)


# =========================================================
# DIAGNOSTIC DISTRIBUTIONS
# =========================================================

cell25_claim_type_distribution = Counter(
    record.get(
        "claim_type"
    )
    for record in cell25_candidate_skeletons
)

cell25_evidence_count_distribution = Counter(
    len(
        record[
            "evidence_ids"
        ]
    )
    for record in cell25_candidate_skeletons
)

cell25_source_type_distribution = Counter(
    source_type
    for record in cell25_candidate_skeletons
    for source_type in record[
        "source_types"
    ]
)


# =========================================================
# VALIDATION CHECKS
# =========================================================

cell25_validation_checks = {
    "record_contract_defined": (
        isinstance(
            cell25_record_contract,
            dict,
        )
    ),

    "no_duplicate_declared_fields": (
        len(
            duplicate_declared_fields
        )
        == 0
    ),

    "3813_candidate_skeletons": (
        len(
            cell25_candidate_skeletons
        )
        == 3813
    ),

    "zero_skeleton_errors": (
        len(
            cell25_skeleton_errors
        )
        == 0
    ),

    "candidate_ids_unique": (
        len(
            set(
                cell25_candidate_ids
            )
        )
        == 3813
    ),

    "seed_ids_unique": (
        len(
            set(
                cell25_seed_ids
            )
        )
        == 3813
    ),

    "all_required_fields_present": (
        cell25_all_required_fields_present
    ),

    "generation_fields_intentionally_empty": (
        cell25_generation_fields_empty
    ),

    "zero_provenance_mismatches": (
        len(
            cell25_provenance_mismatches
        )
        == 0
    ),

    "all_records_have_evidence": (
        all(
            len(
                record[
                    "evidence_ids"
                ]
            )
            >= 1
            for record in (
                cell25_candidate_skeletons
            )
        )
    ),

    "all_records_have_exact_evidence_quotes": (
        all(
            len(
                record[
                    "evidence_quotes"
                ]
            )
            == len(
                record[
                    "evidence_ids"
                ]
            )
            and all(
                isinstance(
                    quote,
                    str,
                )
                and bool(
                    quote.strip()
                )
                for quote in record[
                    "evidence_quotes"
                ]
            )
            for record in (
                cell25_candidate_skeletons
            )
        )
    ),

    "contract_fingerprint_created": (
        len(
            cell25_record_contract_sha256
        )
        == 64
    ),
}


cell25_status = (
    "PASS"
    if all(
        cell25_validation_checks.values()
    )
    else "FAIL"
)


cell25_next_readiness = (
    "READY_FOR_GOLDEN_V2_GENERATION_PROTOCOL_DESIGN"
    if cell25_status == "PASS"
    else "BLOCKED_GOLDEN_V2_RECORD_CONTRACT_REPAIR_REQUIRED"
)


# =========================================================
# OUTPUT
# =========================================================

print(
    "=== GOLDEN V2 RECORD CONTRACT DESIGN ==="
)

print()

print(
    "Contract version:",
    GOLDEN_V2_RECORD_CONTRACT_VERSION,
)

print(
    "Dataset candidate version:",
    GOLDEN_V2_DATASET_CANDIDATE_VERSION,
)

print(
    "Contract SHA-256:",
    cell25_record_contract_sha256,
)

print()

print(
    "Expected candidate population:",
    cell25_record_contract[
        "expected_candidate_population"
    ],
)

print(
    "Candidate skeletons built:",
    len(
        cell25_candidate_skeletons
    ),
)

print(
    "Skeleton errors:",
    len(
        cell25_skeleton_errors
    ),
)

print()

print(
    "FIELD GROUPS:"
)

for (
    group_name,
    fields,
) in cell25_record_contract[
    "field_groups"
].items():

    print(
        " -",
        group_name,
        ":",
        fields,
    )

print()

print(
    "EXPLICITLY DEFERRED FIELDS:"
)

for (
    field_name,
    policy,
) in cell25_record_contract[
    "explicitly_deferred_fields"
].items():

    print(
        " -",
        field_name,
        ":",
        policy,
    )

print()

print(
    "CLAIM TYPE DISTRIBUTION:"
)

print(
    dict(
        sorted(
            cell25_claim_type_distribution.items(),
            key=lambda item: str(
                item[0]
            ),
        )
    )
)

print()

print(
    "EVIDENCE COUNT PER CANDIDATE:"
)

print(
    dict(
        sorted(
            cell25_evidence_count_distribution.items()
        )
    )
)

print()

print(
    "SOURCE TYPE DISTRIBUTION:"
)

print(
    dict(
        sorted(
            cell25_source_type_distribution.items()
        )
    )
)

print()

print(
    "SAMPLE PRE-GENERATION RECORDS:"
)

for sample_index, sample in enumerate(
    cell25_candidate_skeletons[
        :3
    ],
    start=1,
):

    print()

    print(
        f"--- SAMPLE {sample_index} ---"
    )

    print(
        json.dumps(
            sample,
            ensure_ascii=False,
            indent=2,
        )
    )

print()

if cell25_skeleton_errors:

    print(
        "SKELETON ERROR SAMPLE:"
    )

    for error in cell25_skeleton_errors[
        :20
    ]:

        print(
            json.dumps(
                error,
                ensure_ascii=False,
            )
        )

    print()


if cell25_provenance_mismatches:

    print(
        "PROVENANCE MISMATCH SAMPLE:"
    )

    for seed_slot_id in (
        cell25_provenance_mismatches[
            :20
        ]
    ):

        print(
            " -",
            seed_slot_id,
        )

    print()


print(
    "Validation checks:"
)

for (
    check_name,
    check_value,
) in cell25_validation_checks.items():

    print(
        " -",
        check_name,
        ":",
        (
            "PASS"
            if check_value
            else "FAIL"
        ),
    )

print()

print(
    "API calls performed:",
    0,
)

print(
    "Files modified:",
    0,
)

print(
    "Questions generated:",
    0,
)

print(
    "Golden candidate records finalized:",
    0,
)

print(
    "Next readiness:",
    cell25_next_readiness,
)

print(
    "GOLDEN V2 RECORD CONTRACT DESIGN:",
    cell25_status,
)

assert cell25_status == "PASS", (
    "Golden v2 Record Contract design failed."
)
````

<h2>مرحله ۲۶ — طراحی Generation Protocol برای Question و Expected Answer</h2>

<p>
Contract رکورد Golden Dataset v2 برای هر ۳۸۱۳ Seed پذیرفته‌شده با موفقیت
تثبیت شد. تمام فیلدهای هویتی، Claim، Evidence و Source Lineage اکنون پیش
از Generation قابل ساخت و اعتبارسنجی هستند.
</p>

<p>
مرحله جاری قرارداد تولید محتوای ارزیابی را تعریف می‌کند. برای هر Seed باید
یک سؤال مستقل، روشن و قابل پاسخ از Evidence موجود ساخته شود و پاسخ مورد
انتظار نیز فقط بر اساس همان Claim و Evidence تولید شود.
</p>

<p>
Generation اجازه استفاده از دانش خارجی، تکمیل حدسی اطلاعات، تغییر معنای
Claim یا ساخت سؤال فراتر از Evidence را ندارد. سؤال نباید به عباراتی مانند
«طبق متن بالا»، «در این متن» یا Context ارائه‌نشده وابسته باشد.
</p>

<p>
از آنجا که بیشتر Seedها یک Evidence دارند ولی بخشی از آن‌ها دارای چند
Evidence هستند، Protocol باید ورودی چند-Evidence را نیز به‌صورت صریح
پشتیبانی کند.
</p>

<p>
در این مرحله فقط Prompt Contract، Output Schema، Intent Taxonomy،
Validation Rules و Request Fingerprint طراحی و روی ۳۸۱۳ ورودی واقعی
اعتبارسنجی می‌شوند. هیچ درخواست API ارسال نمی‌شود و هیچ Question یا
Answer واقعی تولید نمی‌شود.
</p>

<p>
پس از PASS شدن این مرحله، ابتدا یک Canary کوچک و کنترل‌شده اجرا خواهد شد.
Full Generation تنها در صورتی آغاز می‌شود که Canary از Gateهای ساختاری
و معنایی عبور کند.
</p>

</div>

````python
# Code Cell 26 — Golden v2 Generation Protocol Design
# =========================================================
# Goals:
# - Freeze the Question/Answer generation protocol in memory
# - Define exact model input
# - Define exact JSON output schema
# - Define allowed question intents
# - Define structural validation rules
# - Build deterministic request fingerprints for all 3813 Seeds
# - Select a deterministic, diverse Canary population
#
# Safety:
# - ZERO API calls
# - ZERO file writes
# - ZERO generated Questions
# - ZERO generated Answers
# =========================================================

from collections import Counter, defaultdict
import hashlib
import json
import re


# =========================================================
# REQUIRED STATE FROM CELL 25
# =========================================================

required_cell26_names = [
    "cell25_candidate_skeletons",
    "cell25_record_contract",
    "cell25_record_contract_sha256",
    "GOLDEN_V2_RECORD_CONTRACT_VERSION",
]

missing_cell26_names = [
    name
    for name in required_cell26_names
    if name not in globals()
]

assert not missing_cell26_names, (
    "Missing required Cell 25 state: "
    + ", ".join(missing_cell26_names)
)

assert len(cell25_candidate_skeletons) == 3813


# =========================================================
# GENERATION PROTOCOL IDENTITY
# =========================================================

GOLDEN_V2_GENERATION_PROTOCOL_VERSION = (
    "golden_v2_question_answer_generation_v1"
)

GOLDEN_V2_GENERATION_PROMPT_VERSION = (
    "golden_v2_grounded_qa_prompt_v1"
)

GOLDEN_V2_GENERATION_OUTPUT_SCHEMA_VERSION = (
    "golden_v2_grounded_qa_output_v1"
)


# =========================================================
# ALLOWED QUESTION INTENTS
#
# Intents describe the evaluation target, not the source type.
# =========================================================

GOLDEN_V2_ALLOWED_QUESTION_INTENTS = (
    "DIRECT_FACT",
    "DEFINITION",
    "EXPLANATION",
    "CAUSAL_REASONING",
    "COMPARISON",
    "CONDITION_OR_CONSTRAINT",
    "PROCEDURE_OR_ACTION",
    "RULE_OR_PRINCIPLE",
    "RELATION",
    "CLASSIFICATION",
    "QUANTITATIVE",
)


# =========================================================
# CLAIM TYPE -> PREFERRED INTENT MAPPING
#
# This is guidance for generation, not a forced answer.
# =========================================================

GOLDEN_V2_CLAIM_TYPE_INTENT_GUIDANCE = {
    "DEFINITION": [
        "DEFINITION",
        "DIRECT_FACT",
    ],

    "CAUSAL": [
        "CAUSAL_REASONING",
        "EXPLANATION",
    ],

    "COMPARISON": [
        "COMPARISON",
    ],

    "CONDITION_OR_CONSTRAINT": [
        "CONDITION_OR_CONSTRAINT",
        "RULE_OR_PRINCIPLE",
    ],

    "NUMERIC_OR_QUANTITATIVE": [
        "QUANTITATIVE",
        "DIRECT_FACT",
    ],

    "PROCEDURE_OR_ACTION": [
        "PROCEDURE_OR_ACTION",
    ],

    "RELATION": [
        "RELATION",
        "EXPLANATION",
    ],

    "RULE_OR_PRINCIPLE": [
        "RULE_OR_PRINCIPLE",
        "EXPLANATION",
    ],

    "CLASSIFICATION": [
        "CLASSIFICATION",
        "DIRECT_FACT",
    ],

    "OTHER_ATOMIC": [
        "DIRECT_FACT",
        "EXPLANATION",
    ],

    None: [
        "DIRECT_FACT",
        "EXPLANATION",
    ],
}


# =========================================================
# OUTPUT CONTRACT
#
# Model must return exactly these top-level fields.
# =========================================================

cell26_generation_output_contract = {
    "schema_version": (
        GOLDEN_V2_GENERATION_OUTPUT_SCHEMA_VERSION
    ),

    "required_top_level_fields": [
        "candidate_id",
        "question",
        "expected_answer",
        "question_intent",
    ],

    "additional_top_level_fields_allowed": False,

    "field_rules": {
        "candidate_id": {
            "type": "string",
            "must_equal_input_candidate_id": True,
        },

        "question": {
            "type": "string",
            "must_be_nonempty": True,
            "must_be_standalone": True,
            "must_be_supported_by_claim_and_evidence": True,
            "must_not_require_external_context": True,
        },

        "expected_answer": {
            "type": "string",
            "must_be_nonempty": True,
            "must_be_grounded_in_attached_evidence": True,
            "must_not_add_external_facts": True,
        },

        "question_intent": {
            "type": "enum",
            "allowed_values": list(
                GOLDEN_V2_ALLOWED_QUESTION_INTENTS
            ),
        },
    },
}


# =========================================================
# SYSTEM PROMPT
# =========================================================

GOLDEN_V2_GENERATION_SYSTEM_PROMPT = """
You generate Persian benchmark question-answer records for a strict Golden Dataset.

Use ONLY the supplied claim and attached evidence.
Do not use external knowledge.
Do not invent, infer, complete, or repair facts that are not supported by the supplied material.

Generate exactly ONE independent benchmark question and ONE expected answer.

QUESTION RULES:
- The question must be in Persian.
- It must be standalone and understandable without seeing this prompt.
- It must have an unambiguous scope.
- It must test the supplied claim.
- It must be answerable from the attached evidence.
- It must not refer to "the text above", "this passage", "the supplied evidence", or other hidden context.
- Do not ask a broader question than the evidence can support.
- Prefer a natural question over a templated or awkward question.
- Do not copy the claim mechanically into interrogative form when a clearer natural question is possible.

ANSWER RULES:
- The answer must be in Persian.
- It must directly answer the generated question.
- It must be grounded only in the attached evidence.
- It may paraphrase the evidence without changing its meaning.
- Do not add unsupported facts.
- Keep the answer focused on the evaluation target.

MULTI-EVIDENCE RULE:
- Multiple evidence items may be supplied.
- Use only the evidence needed to support the same claim.
- Do not combine unrelated facts merely because multiple evidence items are present.

INTENT:
Choose exactly one allowed question_intent from the supplied allowed list.

OUTPUT:
Return JSON only.
Return exactly these four top-level fields:
{
  "candidate_id": "...",
  "question": "...",
  "expected_answer": "...",
  "question_intent": "..."
}

Do not expose reasoning.
Do not return markdown.
Do not return explanations outside the JSON.
""".strip()


# =========================================================
# FORBIDDEN CONTEXT-DEPENDENT QUESTION PHRASES
# =========================================================

GOLDEN_V2_FORBIDDEN_QUESTION_PHRASES = (
    "طبق متن بالا",
    "بر اساس متن بالا",
    "در متن بالا",
    "در این متن",
    "طبق این متن",
    "بر اساس این متن",
    "در این بخش",
    "طبق بخش بالا",
    "بر اساس شواهد بالا",
    "در شواهد بالا",
    "طبق شواهد",
    "بر اساس شواهد ارائه‌شده",
)


# =========================================================
# PAYLOAD BUILDER
# =========================================================

def cell26_build_generation_payload(
    skeleton,
):
    claim_type = skeleton.get(
        "claim_type"
    )

    preferred_intents = (
        GOLDEN_V2_CLAIM_TYPE_INTENT_GUIDANCE.get(
            claim_type,
            [
                "DIRECT_FACT",
                "EXPLANATION",
            ],
        )
    )

    evidence = []

    for (
        evidence_id,
        evidence_quote,
        evidence_locator,
        source_id,
        document_id,
    ) in zip(
        skeleton[
            "evidence_ids"
        ],
        skeleton[
            "evidence_quotes"
        ],
        skeleton[
            "evidence_locators"
        ],
        [
            None
        ] * len(
            skeleton[
                "evidence_ids"
            ]
        ),
        [
            None
        ] * len(
            skeleton[
                "evidence_ids"
            ]
        ),
    ):
        evidence.append({
            "evidence_id": (
                evidence_id
            ),

            "quote": (
                evidence_quote
            ),

            "locator": (
                evidence_locator
            ),
        })

    return {
        "candidate_id": (
            skeleton[
                "candidate_id"
            ]
        ),

        "seed_slot_id": (
            skeleton[
                "source_seed_slot_id"
            ]
        ),

        "claim": {
            "claim_id": (
                skeleton[
                    "claim_id"
                ]
            ),

            "text_fa": (
                skeleton[
                    "claim_text_fa"
                ]
            ),

            "claim_type": (
                claim_type
            ),

            "claim_types": (
                skeleton[
                    "claim_types"
                ]
            ),

            "authority_class": (
                skeleton[
                    "authority_class"
                ]
            ),
        },

        "evidence": (
            evidence
        ),

        "source_lineage": {
            "source_ids": (
                skeleton[
                    "source_ids"
                ]
            ),

            "document_ids": (
                skeleton[
                    "document_ids"
                ]
            ),

            "source_types": (
                skeleton[
                    "source_types"
                ]
            ),
        },

        "allowed_question_intents": list(
            GOLDEN_V2_ALLOWED_QUESTION_INTENTS
        ),

        "preferred_question_intents": list(
            preferred_intents
        ),

        "output_schema_version": (
            GOLDEN_V2_GENERATION_OUTPUT_SCHEMA_VERSION
        ),
    }


# =========================================================
# USER PROMPT BUILDER
# =========================================================

def cell26_build_generation_user_prompt(
    skeleton,
):
    payload = (
        cell26_build_generation_payload(
            skeleton
        )
    )

    return (
        json.dumps(
            payload,
            ensure_ascii=False,
            sort_keys=True,
            separators=(
                ",",
                ":",
            ),
        )
        + "\n/no_think"
    )


# =========================================================
# STRUCTURAL OUTPUT VALIDATOR
#
# This validates syntax/shape only.
# Semantic grounding requires a later judge/gate.
# =========================================================

def cell26_validate_generation_output(
    output,
    expected_candidate_id,
):
    errors = []

    if not isinstance(
        output,
        dict,
    ):
        return [
            "OUTPUT_NOT_OBJECT"
        ]

    expected_fields = set(
        cell26_generation_output_contract[
            "required_top_level_fields"
        ]
    )

    actual_fields = set(
        output.keys()
    )

    if actual_fields != expected_fields:
        errors.append(
            "TOP_LEVEL_FIELDS_NOT_EXACT"
        )

    candidate_id = output.get(
        "candidate_id"
    )

    if (
        candidate_id
        != expected_candidate_id
    ):
        errors.append(
            "CANDIDATE_ID_MISMATCH"
        )

    question = output.get(
        "question"
    )

    if (
        not isinstance(
            question,
            str,
        )
        or not question.strip()
    ):
        errors.append(
            "QUESTION_EMPTY_OR_INVALID"
        )

    else:
        normalized_question = (
            " ".join(
                question.split()
            )
        )

        if any(
            phrase in normalized_question
            for phrase in (
                GOLDEN_V2_FORBIDDEN_QUESTION_PHRASES
            )
        ):
            errors.append(
                "QUESTION_DEPENDS_ON_HIDDEN_CONTEXT"
            )

        if len(
            normalized_question
        ) < 5:
            errors.append(
                "QUESTION_TOO_SHORT"
            )

    expected_answer = output.get(
        "expected_answer"
    )

    if (
        not isinstance(
            expected_answer,
            str,
        )
        or not expected_answer.strip()
    ):
        errors.append(
            "EXPECTED_ANSWER_EMPTY_OR_INVALID"
        )

    question_intent = output.get(
        "question_intent"
    )

    if (
        question_intent
        not in GOLDEN_V2_ALLOWED_QUESTION_INTENTS
    ):
        errors.append(
            "QUESTION_INTENT_INVALID"
        )

    return errors


# =========================================================
# REQUEST FINGERPRINT
#
# Fingerprint binds:
# - Record contract
# - Generation protocol
# - System prompt
# - Exact user payload
# =========================================================

def cell26_request_fingerprint(
    skeleton,
):
    user_prompt = (
        cell26_build_generation_user_prompt(
            skeleton
        )
    )

    fingerprint_payload = {
        "record_contract_sha256": (
            cell25_record_contract_sha256
        ),

        "generation_protocol_version": (
            GOLDEN_V2_GENERATION_PROTOCOL_VERSION
        ),

        "prompt_version": (
            GOLDEN_V2_GENERATION_PROMPT_VERSION
        ),

        "output_schema_version": (
            GOLDEN_V2_GENERATION_OUTPUT_SCHEMA_VERSION
        ),

        "system_prompt": (
            GOLDEN_V2_GENERATION_SYSTEM_PROMPT
        ),

        "user_prompt": (
            user_prompt
        ),
    }

    canonical = json.dumps(
        fingerprint_payload,
        ensure_ascii=False,
        sort_keys=True,
        separators=(
            ",",
            ":",
        ),
    )

    return hashlib.sha256(
        canonical.encode(
            "utf-8"
        )
    ).hexdigest()


# =========================================================
# BUILD ALL GENERATION REQUEST DESCRIPTORS
#
# Still ZERO API calls.
# =========================================================

cell26_generation_requests = []

cell26_request_errors = []

for skeleton in (
    cell25_candidate_skeletons
):
    try:
        payload = (
            cell26_build_generation_payload(
                skeleton
            )
        )

        user_prompt = (
            cell26_build_generation_user_prompt(
                skeleton
            )
        )

        request_fingerprint = (
            cell26_request_fingerprint(
                skeleton
            )
        )

        cell26_generation_requests.append({
            "candidate_id": (
                skeleton[
                    "candidate_id"
                ]
            ),

            "source_seed_slot_id": (
                skeleton[
                    "source_seed_slot_id"
                ]
            ),

            "claim_type": (
                skeleton[
                    "claim_type"
                ]
            ),

            "source_types": list(
                skeleton[
                    "source_types"
                ]
            ),

            "evidence_count": len(
                skeleton[
                    "evidence_ids"
                ]
            ),

            "payload": (
                payload
            ),

            "user_prompt": (
                user_prompt
            ),

            "request_fingerprint": (
                request_fingerprint
            ),
        })

    except Exception as exc:
        cell26_request_errors.append({
            "candidate_id": (
                skeleton.get(
                    "candidate_id"
                )
            ),

            "error": (
                repr(
                    exc
                )
            ),
        })


# =========================================================
# REQUEST VALIDATION
# =========================================================

cell26_request_candidate_ids = [
    item[
        "candidate_id"
    ]
    for item in (
        cell26_generation_requests
    )
]

cell26_request_fingerprints = [
    item[
        "request_fingerprint"
    ]
    for item in (
        cell26_generation_requests
    )
]

cell26_payload_candidate_id_mismatches = [
    item[
        "candidate_id"
    ]
    for item in (
        cell26_generation_requests
    )
    if item[
        "payload"
    ][
        "candidate_id"
    ]
    != item[
        "candidate_id"
    ]
]

cell26_empty_evidence_requests = [
    item[
        "candidate_id"
    ]
    for item in (
        cell26_generation_requests
    )
    if not item[
        "payload"
    ][
        "evidence"
    ]
]


# =========================================================
# DETERMINISTIC CANARY SELECTION
#
# Canary goals:
# - cover multiple Claim types
# - cover all source types where possible
# - include multi-evidence cases
# - deterministic across reruns
#
# This only SELECTS records.
# It does NOT execute them.
# =========================================================

CELL26_CANARY_TARGET_SIZE = 12

cell26_requests_sorted = sorted(
    cell26_generation_requests,
    key=lambda item: (
        item[
            "candidate_id"
        ]
    ),
)

cell26_canary_ids = []

cell26_canary_id_set = set()


def cell26_add_canary(
    request,
):
    candidate_id = request[
        "candidate_id"
    ]

    if candidate_id in cell26_canary_id_set:
        return False

    if (
        len(
            cell26_canary_ids
        )
        >= CELL26_CANARY_TARGET_SIZE
    ):
        return False

    cell26_canary_ids.append(
        candidate_id
    )

    cell26_canary_id_set.add(
        candidate_id
    )

    return True


# ---------------------------------------------------------
# First: ensure multi-evidence representation
# ---------------------------------------------------------

multi_evidence_requests = [
    item
    for item in cell26_requests_sorted
    if item[
        "evidence_count"
    ] > 1
]

if multi_evidence_requests:
    cell26_add_canary(
        multi_evidence_requests[0]
    )

max_evidence_request = max(
    cell26_requests_sorted,
    key=lambda item: (
        item[
            "evidence_count"
        ],
        item[
            "candidate_id"
        ],
    ),
)

cell26_add_canary(
    max_evidence_request
)


# ---------------------------------------------------------
# Second: source-type coverage
# ---------------------------------------------------------

for desired_source_type in [
    "book",
    "transcript",
    "website",
]:
    matching = [
        item
        for item in cell26_requests_sorted
        if desired_source_type
        in item[
            "source_types"
        ]
    ]

    if matching:
        cell26_add_canary(
            matching[0]
        )


# ---------------------------------------------------------
# Third: Claim-type coverage
# ---------------------------------------------------------

claim_type_priority = [
    "DEFINITION",
    "RELATION",
    "RULE_OR_PRINCIPLE",
    "CAUSAL",
    "PROCEDURE_OR_ACTION",
    "CONDITION_OR_CONSTRAINT",
    "NUMERIC_OR_QUANTITATIVE",
    "CLASSIFICATION",
    "COMPARISON",
    "OTHER_ATOMIC",
    None,
]

for desired_claim_type in (
    claim_type_priority
):
    matching = [
        item
        for item in cell26_requests_sorted
        if item[
            "claim_type"
        ]
        == desired_claim_type
    ]

    if matching:
        cell26_add_canary(
            matching[0]
        )

    if (
        len(
            cell26_canary_ids
        )
        >= CELL26_CANARY_TARGET_SIZE
    ):
        break


# ---------------------------------------------------------
# Final deterministic fill if still below target
# ---------------------------------------------------------

for item in cell26_requests_sorted:

    if (
        len(
            cell26_canary_ids
        )
        >= CELL26_CANARY_TARGET_SIZE
    ):
        break

    cell26_add_canary(
        item
    )


cell26_canary_requests = [
    item
    for item in cell26_generation_requests
    if item[
        "candidate_id"
    ]
    in cell26_canary_id_set
]

cell26_canary_requests.sort(
    key=lambda item: (
        cell26_canary_ids.index(
            item[
                "candidate_id"
            ]
        )
    )
)


# =========================================================
# CANARY COVERAGE DIAGNOSTICS
# =========================================================

cell26_canary_claim_types = Counter(
    item[
        "claim_type"
    ]
    for item in (
        cell26_canary_requests
    )
)

cell26_canary_source_types = Counter(
    source_type
    for item in (
        cell26_canary_requests
    )
    for source_type in item[
        "source_types"
    ]
)

cell26_canary_evidence_counts = Counter(
    item[
        "evidence_count"
    ]
    for item in (
        cell26_canary_requests
    )
)


# =========================================================
# PROTOCOL FINGERPRINT
# =========================================================

cell26_protocol_manifest = {
    "generation_protocol_version": (
        GOLDEN_V2_GENERATION_PROTOCOL_VERSION
    ),

    "prompt_version": (
        GOLDEN_V2_GENERATION_PROMPT_VERSION
    ),

    "output_schema_version": (
        GOLDEN_V2_GENERATION_OUTPUT_SCHEMA_VERSION
    ),

    "record_contract_sha256": (
        cell25_record_contract_sha256
    ),

    "allowed_question_intents": list(
        GOLDEN_V2_ALLOWED_QUESTION_INTENTS
    ),

    "claim_type_intent_guidance": {
        (
            "__NONE__"
            if key is None
            else key
        ): value
        for key, value in GOLDEN_V2_CLAIM_TYPE_INTENT_GUIDANCE.items()
    },

    "output_contract": (
        cell26_generation_output_contract
    ),

    "system_prompt": (
        GOLDEN_V2_GENERATION_SYSTEM_PROMPT
    ),

    "forbidden_question_phrases": list(
        GOLDEN_V2_FORBIDDEN_QUESTION_PHRASES
    ),

    "canary_target_size": (
        CELL26_CANARY_TARGET_SIZE
    ),

    "canary_candidate_ids": (
        cell26_canary_ids
    ),
}

cell26_protocol_canonical_json = (
    json.dumps(
        cell26_protocol_manifest,
        ensure_ascii=False,
        sort_keys=True,
        separators=(
            ",",
            ":",
        ),
    )
)

cell26_protocol_sha256 = (
    hashlib.sha256(
        cell26_protocol_canonical_json.encode(
            "utf-8"
        )
    ).hexdigest()
)


# =========================================================
# VALIDATION CHECKS
# =========================================================

cell26_validation_checks = {
    "3813_generation_requests_built": (
        len(
            cell26_generation_requests
        )
        == 3813
    ),

    "zero_request_build_errors": (
        len(
            cell26_request_errors
        )
        == 0
    ),

    "request_candidate_ids_unique": (
        len(
            set(
                cell26_request_candidate_ids
            )
        )
        == 3813
    ),

    "request_fingerprints_unique": (
        len(
            set(
                cell26_request_fingerprints
            )
        )
        == 3813
    ),

    "zero_payload_candidate_id_mismatches": (
        len(
            cell26_payload_candidate_id_mismatches
        )
        == 0
    ),

    "zero_empty_evidence_requests": (
        len(
            cell26_empty_evidence_requests
        )
        == 0
    ),

    "allowed_intents_nonempty": (
        len(
            GOLDEN_V2_ALLOWED_QUESTION_INTENTS
        )
        > 0
    ),

    "all_claim_types_have_intent_guidance": (
        all(
            skeleton.get(
                "claim_type"
            )
            in GOLDEN_V2_CLAIM_TYPE_INTENT_GUIDANCE
            for skeleton in (
                cell25_candidate_skeletons
            )
        )
    ),

    "canary_population_exact_12": (
        len(
            cell26_canary_requests
        )
        == CELL26_CANARY_TARGET_SIZE
    ),

    "canary_contains_multi_evidence": (
        any(
            item[
                "evidence_count"
            ]
            > 1
            for item in (
                cell26_canary_requests
            )
        )
    ),

    "canary_covers_book": (
        "book"
        in cell26_canary_source_types
    ),

    "canary_covers_transcript": (
        "transcript"
        in cell26_canary_source_types
    ),

    "canary_covers_website": (
        "website"
        in cell26_canary_source_types
    ),

    "protocol_fingerprint_created": (
        len(
            cell26_protocol_sha256
        )
        == 64
    ),
}


cell26_status = (
    "PASS"
    if all(
        cell26_validation_checks.values()
    )
    else "FAIL"
)


cell26_next_readiness = (
    "READY_FOR_GOLDEN_V2_GENERATION_CANARY"
    if cell26_status == "PASS"
    else "BLOCKED_GENERATION_PROTOCOL_REPAIR_REQUIRED"
)


# =========================================================
# OUTPUT
# =========================================================

print(
    "=== GOLDEN V2 GENERATION PROTOCOL DESIGN ==="
)

print()

print(
    "Generation protocol version:",
    GOLDEN_V2_GENERATION_PROTOCOL_VERSION,
)

print(
    "Prompt version:",
    GOLDEN_V2_GENERATION_PROMPT_VERSION,
)

print(
    "Output schema version:",
    GOLDEN_V2_GENERATION_OUTPUT_SCHEMA_VERSION,
)

print(
    "Protocol SHA-256:",
    cell26_protocol_sha256,
)

print()

print(
    "Generation requests prepared:",
    len(
        cell26_generation_requests
    ),
)

print(
    "Request build errors:",
    len(
        cell26_request_errors
    ),
)

print(
    "Unique request fingerprints:",
    len(
        set(
            cell26_request_fingerprints
        )
    ),
)

print()

print(
    "Allowed question intents:"
)

for intent in (
    GOLDEN_V2_ALLOWED_QUESTION_INTENTS
):
    print(
        " -",
        intent,
    )

print()

print(
    "Canary target:",
    CELL26_CANARY_TARGET_SIZE,
)

print(
    "Canary selected:",
    len(
        cell26_canary_requests
    ),
)

print(
    "Canary Claim types:",
    dict(
        cell26_canary_claim_types
    ),
)

print(
    "Canary Source types:",
    dict(
        cell26_canary_source_types
    ),
)

print(
    "Canary Evidence counts:",
    dict(
        sorted(
            cell26_canary_evidence_counts.items()
        )
    ),
)

print()

print(
    "CANARY RECORDS:"
)

for index, item in enumerate(
    cell26_canary_requests,
    start=1,
):
    print(
        f" {index:02d}.",
        item[
            "candidate_id"
        ],
        "| claim_type:",
        item[
            "claim_type"
        ],
        "| source_types:",
        item[
            "source_types"
        ],
        "| evidence_count:",
        item[
            "evidence_count"
        ],
    )

print()

print(
    "SAMPLE GENERATION PAYLOAD:"
)

print(
    json.dumps(
        cell26_generation_requests[
            0
        ][
            "payload"
        ],
        ensure_ascii=False,
        indent=2,
    )
)

print()

print(
    "Validation checks:"
)

for (
    check_name,
    check_value,
) in cell26_validation_checks.items():

    print(
        " -",
        check_name,
        ":",
        (
            "PASS"
            if check_value
            else "FAIL"
        ),
    )

print()

print(
    "API calls performed:",
    0,
)

print(
    "Files modified:",
    0,
)

print(
    "Questions generated:",
    0,
)

print(
    "Answers generated:",
    0,
)

print(
    "Full Generation started:",
    False,
)

print(
    "Next readiness:",
    cell26_next_readiness,
)

print(
    "GOLDEN V2 GENERATION PROTOCOL DESIGN:",
    cell26_status,
)

assert cell26_status == "PASS", (
    "Golden v2 Generation Protocol design failed."
)
````

<h2>مرحله ۲۸ — Semantic Audit دوازده خروجی Canary</h2>

<p>
Canary تولید Golden v2 از نظر اجرای API و ساختار JSON برای هر ۱۲ نمونه
موفق بود. با این حال Structural PASS به‌تنهایی برای ورود یک رکورد به
Golden Dataset کافی نیست.
</p>

<p>
در این مرحله Claim اصلی، Evidence دقیق، Question تولیدشده و Expected Answer
برای هر Canary کنار یکدیگر قرار می‌گیرند تا کیفیت معنایی خروجی قابل بررسی
باشد.
</p>

<p>
کنترل‌های deterministic شامل حفظ Candidate ID، وجود Question و Answer،
عدم وابستگی سؤال به Context مخفی، تطابق Intent با مقادیر مجاز، شباهت
متنی Answer با Evidence و تشخیص برخی الگوهای مشکوک است.
</p>

<p>
این کنترل‌های ماشینی جای Semantic Judgment را نمی‌گیرند؛ بنابراین نتیجه
این مرحله به‌صورت Audit Packet ارائه می‌شود و Full Generation همچنان
متوقف باقی می‌ماند تا کیفیت واقعی Canary مشخص شود.
</p>

<p>
در این مرحله هیچ API فراخوانی نمی‌شود، هیچ Artifact نوشته نمی‌شود و هیچ
رکورد Golden نهایی نمی‌شود.
</p>

````python
# Code Cell 28 — Canary Semantic Audit Packet
# =========================================================
# Goals:
# - Join each generated Canary with its exact Claim/Evidence
# - Run deterministic semantic-risk diagnostics
# - Print complete audit packets for all 12 Canary records
# - DO NOT automatically declare semantic PASS
#
# Safety:
# - ZERO API calls
# - ZERO file writes
# - ZERO Full Generation
# =========================================================

from collections import Counter
import json
import re


# =========================================================
# REQUIRED STATE
# =========================================================

required_cell28_names = [
    "cell27_canary_results",
    "cell25_candidate_skeletons",
    "GOLDEN_V2_ALLOWED_QUESTION_INTENTS",
    "GOLDEN_V2_FORBIDDEN_QUESTION_PHRASES",
]

missing_cell28_names = [
    name
    for name in required_cell28_names
    if name not in globals()
]

assert not missing_cell28_names, (
    "Missing required Canary state: "
    + ", ".join(missing_cell28_names)
)

assert len(cell27_canary_results) == 12


# =========================================================
# TEXT NORMALIZATION
# =========================================================

def cell28_normalize_text(text):
    if not isinstance(text, str):
        return ""

    text = text.replace(
        "\u200c",
        " ",
    )

    text = re.sub(
        r"[^\w\u0600-\u06FF]+",
        " ",
        text,
    )

    return " ".join(
        text.lower().split()
    )


def cell28_tokens(text):
    normalized = (
        cell28_normalize_text(
            text
        )
    )

    return {
        token
        for token in normalized.split()
        if len(token) >= 2
    }


def cell28_overlap_ratio(
    source_text,
    target_text,
):
    source_tokens = (
        cell28_tokens(
            source_text
        )
    )

    target_tokens = (
        cell28_tokens(
            target_text
        )
    )

    if not target_tokens:
        return 0.0

    return (
        len(
            source_tokens
            & target_tokens
        )
        /
        len(
            target_tokens
        )
    )


# =========================================================
# LOOKUP ORIGINAL SKELETONS
# =========================================================

cell28_skeleton_by_candidate_id = {
    record[
        "candidate_id"
    ]: record
    for record in cell25_candidate_skeletons
}

assert len(
    cell28_skeleton_by_candidate_id
) == 3813


# =========================================================
# DIAGNOSTIC RULES
#
# These are risk flags, NOT final semantic judgments.
# =========================================================

cell28_audit_packets = []

cell28_build_errors = []


for result in cell27_canary_results:

    candidate_id = result.get(
        "candidate_id"
    )

    skeleton = (
        cell28_skeleton_by_candidate_id.get(
            candidate_id
        )
    )

    parsed = result.get(
        "parsed_json"
    )

    if skeleton is None:

        cell28_build_errors.append({
            "candidate_id": (
                candidate_id
            ),
            "error": (
                "SKELETON_NOT_FOUND"
            ),
        })

        continue

    if not isinstance(
        parsed,
        dict,
    ):

        cell28_build_errors.append({
            "candidate_id": (
                candidate_id
            ),
            "error": (
                "PARSED_GENERATION_NOT_OBJECT"
            ),
        })

        continue

    claim_text = skeleton.get(
        "claim_text_fa",
        "",
    )

    evidence_quotes = skeleton.get(
        "evidence_quotes",
        [],
    )

    combined_evidence = "\n".join(
        evidence_quotes
    )

    question = parsed.get(
        "question",
        "",
    )

    expected_answer = parsed.get(
        "expected_answer",
        "",
    )

    question_intent = parsed.get(
        "question_intent"
    )

    risk_flags = []

    # -----------------------------------------------------
    # Candidate identity
    # -----------------------------------------------------

    if (
        parsed.get(
            "candidate_id"
        )
        != candidate_id
    ):
        risk_flags.append(
            "CANDIDATE_ID_MISMATCH"
        )

    # -----------------------------------------------------
    # Hidden-context dependency
    # -----------------------------------------------------

    normalized_question = (
        " ".join(
            question.split()
        )
        if isinstance(
            question,
            str,
        )
        else ""
    )

    if any(
        phrase
        in normalized_question
        for phrase in (
            GOLDEN_V2_FORBIDDEN_QUESTION_PHRASES
        )
    ):
        risk_flags.append(
            "QUESTION_HIDDEN_CONTEXT_DEPENDENCY"
        )

    # -----------------------------------------------------
    # Intent
    # -----------------------------------------------------

    if (
        question_intent
        not in GOLDEN_V2_ALLOWED_QUESTION_INTENTS
    ):
        risk_flags.append(
            "INVALID_QUESTION_INTENT"
        )

    # -----------------------------------------------------
    # Basic language / punctuation
    # -----------------------------------------------------

    if not isinstance(
        question,
        str,
    ) or not question.strip():

        risk_flags.append(
            "EMPTY_QUESTION"
        )

    elif not (
        question.rstrip().endswith(
            "؟"
        )
        or question.rstrip().endswith(
            "?"
        )
    ):
        risk_flags.append(
            "QUESTION_WITHOUT_QUESTION_MARK"
        )

    if not isinstance(
        expected_answer,
        str,
    ) or not expected_answer.strip():

        risk_flags.append(
            "EMPTY_EXPECTED_ANSWER"
        )

    # -----------------------------------------------------
    # Lexical grounding diagnostics
    #
    # Low overlap is NOT automatically hallucination.
    # It is only a review signal.
    # -----------------------------------------------------

    answer_evidence_overlap = (
        cell28_overlap_ratio(
            combined_evidence,
            expected_answer,
        )
    )

    answer_claim_overlap = (
        cell28_overlap_ratio(
            claim_text,
            expected_answer,
        )
    )

    question_claim_overlap = (
        cell28_overlap_ratio(
            claim_text,
            question,
        )
    )

    if (
        answer_evidence_overlap
        < 0.25
        and
        answer_claim_overlap
        < 0.25
    ):
        risk_flags.append(
            "LOW_ANSWER_GROUNDING_OVERLAP"
        )

    if question_claim_overlap < 0.10:
        risk_flags.append(
            "LOW_QUESTION_CLAIM_OVERLAP"
        )

    # -----------------------------------------------------
    # Suspicious answer expansion
    # -----------------------------------------------------

    answer_token_count = len(
        cell28_tokens(
            expected_answer
        )
    )

    evidence_token_count = len(
        cell28_tokens(
            combined_evidence
        )
    )

    if (
        evidence_token_count > 0
        and
        answer_token_count
        > evidence_token_count * 1.5
    ):
        risk_flags.append(
            "ANSWER_LONGER_THAN_EVIDENCE_SIGNAL"
        )

    # -----------------------------------------------------
    # Claim/evidence relationship diagnostic
    # -----------------------------------------------------

    claim_evidence_overlap = (
        cell28_overlap_ratio(
            combined_evidence,
            claim_text,
        )
    )

    if claim_evidence_overlap < 0.20:
        risk_flags.append(
            "LOW_CLAIM_EVIDENCE_OVERLAP"
        )

    # -----------------------------------------------------
    # Packet
    # -----------------------------------------------------

    cell28_audit_packets.append({
        "candidate_id": (
            candidate_id
        ),

        "source_seed_slot_id": (
            skeleton.get(
                "source_seed_slot_id"
            )
        ),

        "claim_type": (
            skeleton.get(
                "claim_type"
            )
        ),

        "claim_types": (
            skeleton.get(
                "claim_types"
            )
        ),

        "authority_class": (
            skeleton.get(
                "authority_class"
            )
        ),

        "source_types": (
            skeleton.get(
                "source_types"
            )
        ),

        "source_ids": (
            skeleton.get(
                "source_ids"
            )
        ),

        "document_ids": (
            skeleton.get(
                "document_ids"
            )
        ),

        "claim_text_fa": (
            claim_text
        ),

        "evidence_ids": (
            skeleton.get(
                "evidence_ids"
            )
        ),

        "evidence_quotes": (
            evidence_quotes
        ),

        "question": (
            question
        ),

        "expected_answer": (
            expected_answer
        ),

        "question_intent": (
            question_intent
        ),

        "diagnostics": {
            "answer_evidence_overlap": round(
                answer_evidence_overlap,
                4,
            ),

            "answer_claim_overlap": round(
                answer_claim_overlap,
                4,
            ),

            "question_claim_overlap": round(
                question_claim_overlap,
                4,
            ),

            "claim_evidence_overlap": round(
                claim_evidence_overlap,
                4,
            ),

            "risk_flags": (
                risk_flags
            ),

            "risk_flag_count": len(
                risk_flags
            ),
        },

        "semantic_decision": (
            "PENDING_AUDIT"
        ),
    })


# =========================================================
# AGGREGATE DIAGNOSTICS
# =========================================================

cell28_risk_flag_counts = Counter(
    flag
    for packet in cell28_audit_packets
    for flag in packet[
        "diagnostics"
    ][
        "risk_flags"
    ]
)

cell28_zero_flag_count = sum(
    packet[
        "diagnostics"
    ][
        "risk_flag_count"
    ]
    == 0
    for packet in cell28_audit_packets
)

cell28_flagged_count = (
    len(
        cell28_audit_packets
    )
    - cell28_zero_flag_count
)


# =========================================================
# STRUCTURAL AUDIT-PACKET GATE
#
# Important:
# This gate validates the AUDIT PACKET construction.
# It does NOT approve semantic quality.
# =========================================================

cell28_validation_checks = {
    "12_audit_packets_built": (
        len(
            cell28_audit_packets
        )
        == 12
    ),

    "zero_packet_build_errors": (
        len(
            cell28_build_errors
        )
        == 0
    ),

    "all_have_claim_text": (
        all(
            bool(
                packet[
                    "claim_text_fa"
                ].strip()
            )
            for packet in (
                cell28_audit_packets
            )
        )
    ),

    "all_have_evidence": (
        all(
            len(
                packet[
                    "evidence_quotes"
                ]
            )
            >= 1
            for packet in (
                cell28_audit_packets
            )
        )
    ),

    "all_have_generated_question": (
        all(
            isinstance(
                packet[
                    "question"
                ],
                str,
            )
            and bool(
                packet[
                    "question"
                ].strip()
            )
            for packet in (
                cell28_audit_packets
            )
        )
    ),

    "all_have_generated_answer": (
        all(
            isinstance(
                packet[
                    "expected_answer"
                ],
                str,
            )
            and bool(
                packet[
                    "expected_answer"
                ].strip()
            )
            for packet in (
                cell28_audit_packets
            )
        )
    ),

    "all_semantic_decisions_pending": (
        all(
            packet[
                "semantic_decision"
            ]
            == "PENDING_AUDIT"
            for packet in (
                cell28_audit_packets
            )
        )
    ),
}


cell28_status = (
    "PASS"
    if all(
        cell28_validation_checks.values()
    )
    else "FAIL"
)


cell28_next_readiness = (
    "READY_FOR_CANARY_SEMANTIC_ADJUDICATION"
    if cell28_status == "PASS"
    else "BLOCKED_CANARY_AUDIT_PACKET_REPAIR_REQUIRED"
)


# =========================================================
# OUTPUT
# =========================================================

print(
    "=== GOLDEN V2 CANARY SEMANTIC AUDIT PACKET ==="
)

print()

print(
    "Audit packets:",
    len(
        cell28_audit_packets
    ),
)

print(
    "Packet build errors:",
    len(
        cell28_build_errors
    ),
)

print(
    "Zero-risk-flag packets:",
    cell28_zero_flag_count,
)

print(
    "Flagged packets:",
    cell28_flagged_count,
)

print(
    "Risk flag distribution:",
    dict(
        cell28_risk_flag_counts
    ),
)

print()

print(
    "IMPORTANT:"
)

print(
    "Risk flags are diagnostics only; "
    "they are NOT semantic verdicts."
)

print(
    "Full Generation remains disabled."
)

print()


# =========================================================
# PRINT ALL 12 COMPLETE AUDIT PACKETS
# =========================================================

for index, packet in enumerate(
    cell28_audit_packets,
    start=1,
):

    print(
        "=" * 88
    )

    print(
        f"CANARY SEMANTIC AUDIT {index:02d}"
    )

    print(
        "=" * 88
    )

    print(
        "Candidate:",
        packet[
            "candidate_id"
        ],
    )

    print(
        "Seed:",
        packet[
            "source_seed_slot_id"
        ],
    )

    print(
        "Claim type:",
        packet[
            "claim_type"
        ],
    )

    print(
        "Authority:",
        packet[
            "authority_class"
        ],
    )

    print(
        "Source types:",
        packet[
            "source_types"
        ],
    )

    print()

    print(
        "CLAIM:"
    )

    print(
        packet[
            "claim_text_fa"
        ]
    )

    print()

    print(
        "EVIDENCE:"
    )

    for evidence_index, (
        evidence_id,
        evidence_quote,
    ) in enumerate(
        zip(
            packet[
                "evidence_ids"
            ],
            packet[
                "evidence_quotes"
            ],
        ),
        start=1,
    ):

        print(
            f"[{evidence_index}]",
            evidence_id,
        )

        print(
            evidence_quote
        )

        print()

    print(
        "GENERATED QUESTION:"
    )

    print(
        packet[
            "question"
        ]
    )

    print()

    print(
        "EXPECTED ANSWER:"
    )

    print(
        packet[
            "expected_answer"
        ]
    )

    print()

    print(
        "QUESTION INTENT:",
        packet[
            "question_intent"
        ],
    )

    print()

    print(
        "DIAGNOSTICS:"
    )

    print(
        json.dumps(
            packet[
                "diagnostics"
            ],
            ensure_ascii=False,
            indent=2,
        )
    )

    print()

    print(
        "SEMANTIC DECISION:",
        packet[
            "semantic_decision"
        ],
    )

    print()


# =========================================================
# VALIDATION SUMMARY
# =========================================================

print(
    "=" * 88
)

print(
    "Validation checks:"
)

for (
    check_name,
    check_value,
) in cell28_validation_checks.items():

    print(
        " -",
        check_name,
        ":",
        (
            "PASS"
            if check_value
            else "FAIL"
        ),
    )

print()

print(
    "API calls performed:",
    0,
)

print(
    "Files modified:",
    0,
)

print(
    "Full Generation started:",
    False,
)

print(
    "Semantic Canary records approved:",
    0,
)

print(
    "Next readiness:",
    cell28_next_readiness,
)

print(
    "GOLDEN V2 CANARY SEMANTIC AUDIT PACKET:",
    cell28_status,
)

assert cell28_status == "PASS", (
    "Canary Semantic Audit Packet construction failed."
)
````

<h2>مرحله ۲۹ — Adjudication معنایی Canary و تشخیص محل خطا</h2>

<p>
Audit دوازده Canary نشان داد که موفقیت ساختاری Generation به‌تنهایی برای
تأیید Protocol کافی نیست. در این مرحله هر Canary بر اساس Claim، Evidence،
Question و Expected Answer به‌صورت صریح adjudicate می‌شود.
</p>

<p>
هدف اصلی تفکیک دو نوع خطاست: خطای Generation، زمانی که مدل به‌جای Claim
هدف، بخش دیگری از Evidence را به سؤال تبدیل کرده است؛ و خطای Upstream
Claim–Evidence، زمانی که Evidence متصل‌شده به Seed به‌تنهایی تمام معنای
Claim را پشتیبانی نمی‌کند.
</p>

<p>
Canaryهای سالم با ACCEPT_GENERATION مشخص می‌شوند. موارد دارای مشکل
Claim–Evidence با HOLD_UPSTREAM_EVIDENCE و موارد دارای سؤال یا پاسخ
نامنطبق با Claim با REJECT_GENERATION مشخص می‌شوند.
</p>

<p>
این adjudication فقط برای Canary دوازده‌تایی است و هیچ تغییری در ۳۸۱۳
Candidate، Claim Registry، Evidence Registry یا Artifactهای Frozen ایجاد
نمی‌کند. Full Generation همچنان غیرفعال باقی می‌ماند.

<h2>مرحله ۳۰ — طراحی Generation Protocol v2 با Claim-Target Enforcement</h2>

<p>
Adjudication دوازده Canary نشان داد که Generation v1 در ۹ مورد سالم بود،
اما یک مورد سؤال را از بخش دیگری از Evidence ساخت و دو مورد نیز دارای
ناهماهنگی میان Claim و Evidence بودند.
</p>

<p>
در Protocol v2 نقش Claim و Evidence به‌صورت صریح تفکیک می‌شود:
Claim هدف ارزیابی را تعیین می‌کند و Evidence مرز اطلاعات مجاز برای سؤال
و پاسخ را مشخص می‌کند. مدل فقط زمانی اجازه تولید Question و Expected Answer
دارد که Evidence واقعاً Claim هدف را پشتیبانی کند.
</p>

<p>
اگر Evidence برای پشتیبانی Claim کافی نباشد، مدل نباید Claim را صرفاً
تکرار کند یا اطلاعات ناقص را تکمیل کند. در چنین حالتی باید خروجی
UNSUPPORTED_CLAIM تولید شود تا رکورد برای بررسی Upstream نگه داشته شود.
</p>

<p>
همچنین سؤال تولیدشده باید خود Claim هدف را بسنجد. استفاده از حقیقت دیگری
که در Evidence وجود دارد ولی جزو Claim هدف نیست، به‌عنوان
QUESTION_TARGET_MISMATCH در نظر گرفته می‌شود.
</p>

<p>
این مرحله فقط Protocol v2، Schema جدید و Request Fingerprintهای جدید را
در حافظه می‌سازد. هیچ API فراخوانی نمی‌شود و Full Generation همچنان
غیرفعال باقی می‌ماند. پس از PASS، Canary v2 روی همان ۱۲ مورد اجرا خواهد شد
تا امکان مقایسه مستقیم با Generation v1 وجود داشته باشد.
</p>

</div>

````python
# Code Cell 30 — Generation Protocol v2 Claim-Target Enforcement
# =========================================================
# Goals:
# - Fix the observed v1 generation failure
# - Explicitly separate:
#       Claim = evaluation target
#       Evidence = permitted factual support
# - Allow UNSUPPORTED_CLAIM instead of forcing bad Q/A
# - Build deterministic v2 requests for all 3813 candidates
# - Reuse the SAME 12 Canary candidate IDs for direct comparison
#
# Safety:
# - ZERO API calls
# - ZERO file writes
# - ZERO generated questions
# - Full Generation remains disabled
# =========================================================

import hashlib
import json


# =========================================================
# REQUIRED STATE
# =========================================================

required_cell30_names = [
    "cell25_candidate_skeletons",
    "cell25_record_contract_sha256",
    "cell26_canary_requests",
    "cell29_adjudicated_records",
    "GOLDEN_V2_ALLOWED_QUESTION_INTENTS",
]

missing_cell30_names = [
    name
    for name in required_cell30_names
    if name not in globals()
]

assert not missing_cell30_names, (
    "Missing required state: "
    + ", ".join(missing_cell30_names)
)

assert len(cell25_candidate_skeletons) == 3813
assert len(cell26_canary_requests) == 12
assert len(cell29_adjudicated_records) == 12


# =========================================================
# V2 PROTOCOL IDENTITY
# =========================================================

GOLDEN_V2_GENERATION_PROTOCOL_VERSION_V2 = (
    "golden_v2_question_answer_generation_v2"
)

GOLDEN_V2_GENERATION_PROMPT_VERSION_V2 = (
    "golden_v2_claim_target_grounded_qa_prompt_v2"
)

GOLDEN_V2_GENERATION_OUTPUT_SCHEMA_VERSION_V2 = (
    "golden_v2_claim_target_grounded_qa_output_v2"
)


# =========================================================
# GENERATION DECISIONS
# =========================================================

GOLDEN_V2_GENERATION_DECISIONS_V2 = (
    "GENERATE_QA",
    "UNSUPPORTED_CLAIM",
)


# =========================================================
# V2 OUTPUT CONTRACT
# =========================================================

cell30_generation_output_contract_v2 = {
    "schema_version": (
        GOLDEN_V2_GENERATION_OUTPUT_SCHEMA_VERSION_V2
    ),

    "required_top_level_fields": [
        "candidate_id",
        "generation_decision",
        "question",
        "expected_answer",
        "question_intent",
        "decision_reason",
    ],

    "additional_top_level_fields_allowed": False,

    "decision_rules": {
        "GENERATE_QA": {
            "question": "NONEMPTY_STRING",
            "expected_answer": "NONEMPTY_STRING",
            "question_intent": (
                "ONE_ALLOWED_INTENT"
            ),
            "decision_reason": (
                "SHORT_GROUNDING_CONFIRMATION"
            ),
        },

        "UNSUPPORTED_CLAIM": {
            "question": None,
            "expected_answer": None,
            "question_intent": None,
            "decision_reason": (
                "SHORT_EXPLANATION_OF_MISSING_SUPPORT"
            ),
        },
    },
}


# =========================================================
# V2 SYSTEM PROMPT
# =========================================================

GOLDEN_V2_GENERATION_SYSTEM_PROMPT_V2 = """
You generate Persian benchmark question-answer records for a strict evidence-grounded Golden Dataset.

There are TWO different roles in the input:

1. CLAIM = THE EVALUATION TARGET.
2. EVIDENCE = THE ONLY PERMITTED FACTUAL SUPPORT.

You MUST NOT choose an arbitrary fact from the evidence.
You MUST evaluate and generate specifically for the supplied CLAIM.

FIRST, silently determine whether the supplied evidence supports the substantive meaning of the CLAIM.

If the evidence does NOT adequately support the CLAIM:
- Do NOT generate a question.
- Do NOT repeat the unsupported claim as an answer.
- Do NOT repair or complete the evidence using external knowledge.
- Return generation_decision = "UNSUPPORTED_CLAIM".
- Set question, expected_answer, and question_intent to null.
- Give a short Persian decision_reason explaining what support is missing.

If the evidence DOES support the CLAIM:
- Return generation_decision = "GENERATE_QA".
- Generate exactly ONE Persian benchmark question.
- The question MUST test the supplied CLAIM itself.
- The expected answer MUST answer that question using ONLY the supplied evidence.

CRITICAL CLAIM-TARGET RULE:
A fact may appear in the evidence but still be outside the target CLAIM.
Do NOT generate a question about such a fact.
The generated question must evaluate the semantic content of the supplied CLAIM.

QUESTION RULES:
- Persian.
- Standalone.
- Natural and unambiguous.
- Directly targets the CLAIM.
- Answerable from the supplied EVIDENCE.
- Must not require hidden context.
- Must not say "طبق متن بالا", "در این متن", "طبق شواهد", or equivalent.
- Must not broaden the CLAIM beyond what the evidence supports.
- Must not replace the target CLAIM with another interesting fact from the evidence.

ANSWER RULES:
- Persian.
- Directly answers the generated question.
- Supported by the supplied EVIDENCE.
- No external knowledge.
- No invented details.
- No unsupported completion of the CLAIM.
- Paraphrasing is allowed only when meaning is preserved.

MULTI-EVIDENCE RULE:
Multiple evidence items may be supplied.
Use them only to determine whether and how the SAME target CLAIM is supported.
Do not switch the evaluation target to another fact merely because it appears in one evidence item.

INTENT:
When generation_decision is "GENERATE_QA", choose exactly one allowed question_intent.
When generation_decision is "UNSUPPORTED_CLAIM", question_intent must be null.

OUTPUT:
Return JSON only.
Return exactly these six top-level fields:

{
  "candidate_id": "...",
  "generation_decision": "GENERATE_QA or UNSUPPORTED_CLAIM",
  "question": "... or null",
  "expected_answer": "... or null",
  "question_intent": "... or null",
  "decision_reason": "..."
}

Do not expose chain-of-thought.
Do not return markdown.
Do not return any text outside the JSON.
""".strip()


# =========================================================
# V2 PAYLOAD BUILDER
# =========================================================

def cell30_build_generation_payload_v2(
    skeleton,
):

    claim_type = skeleton.get(
        "claim_type"
    )

    preferred_intents = (
        GOLDEN_V2_CLAIM_TYPE_INTENT_GUIDANCE.get(
            claim_type,
            [
                "DIRECT_FACT",
                "EXPLANATION",
            ],
        )
    )

    evidence = []

    for (
        evidence_id,
        evidence_quote,
        evidence_locator,
    ) in zip(
        skeleton[
            "evidence_ids"
        ],
        skeleton[
            "evidence_quotes"
        ],
        skeleton[
            "evidence_locators"
        ],
    ):

        evidence.append({
            "evidence_id": (
                evidence_id
            ),

            "quote": (
                evidence_quote
            ),

            "locator": (
                evidence_locator
            ),
        })

    return {
        "candidate_id": (
            skeleton[
                "candidate_id"
            ]
        ),

        "seed_slot_id": (
            skeleton[
                "source_seed_slot_id"
            ]
        ),

        "evaluation_target_claim": {
            "claim_id": (
                skeleton[
                    "claim_id"
                ]
            ),

            "text_fa": (
                skeleton[
                    "claim_text_fa"
                ]
            ),

            "claim_type": (
                claim_type
            ),

            "claim_types": (
                skeleton[
                    "claim_types"
                ]
            ),

            "authority_class": (
                skeleton[
                    "authority_class"
                ]
            ),
        },

        "permitted_evidence": (
            evidence
        ),

        "source_lineage": {
            "source_ids": (
                skeleton[
                    "source_ids"
                ]
            ),

            "document_ids": (
                skeleton[
                    "document_ids"
                ]
            ),

            "source_types": (
                skeleton[
                    "source_types"
                ]
            ),
        },

        "allowed_generation_decisions": list(
            GOLDEN_V2_GENERATION_DECISIONS_V2
        ),

        "allowed_question_intents": list(
            GOLDEN_V2_ALLOWED_QUESTION_INTENTS
        ),

        "preferred_question_intents": list(
            preferred_intents
        ),

        "output_schema_version": (
            GOLDEN_V2_GENERATION_OUTPUT_SCHEMA_VERSION_V2
        ),
    }


# =========================================================
# V2 USER PROMPT
# =========================================================

def cell30_build_generation_user_prompt_v2(
    skeleton,
):

    payload = (
        cell30_build_generation_payload_v2(
            skeleton
        )
    )

    return (
        json.dumps(
            payload,
            ensure_ascii=False,
            sort_keys=True,
            separators=(
                ",",
                ":",
            ),
        )
        + "\n/no_think"
    )


# =========================================================
# V2 STRUCTURAL VALIDATOR
# =========================================================

def cell30_validate_generation_output_v2(
    output,
    expected_candidate_id,
):

    errors = []

    if not isinstance(
        output,
        dict,
    ):
        return [
            "OUTPUT_NOT_OBJECT"
        ]

    expected_fields = {
        "candidate_id",
        "generation_decision",
        "question",
        "expected_answer",
        "question_intent",
        "decision_reason",
    }

    if set(
        output.keys()
    ) != expected_fields:

        errors.append(
            "TOP_LEVEL_FIELDS_NOT_EXACT"
        )

    if (
        output.get(
            "candidate_id"
        )
        != expected_candidate_id
    ):
        errors.append(
            "CANDIDATE_ID_MISMATCH"
        )

    generation_decision = (
        output.get(
            "generation_decision"
        )
    )

    if (
        generation_decision
        not in GOLDEN_V2_GENERATION_DECISIONS_V2
    ):
        errors.append(
            "INVALID_GENERATION_DECISION"
        )

        return errors

    decision_reason = output.get(
        "decision_reason"
    )

    if (
        not isinstance(
            decision_reason,
            str,
        )
        or not decision_reason.strip()
    ):
        errors.append(
            "DECISION_REASON_EMPTY"
        )

    question = output.get(
        "question"
    )

    expected_answer = output.get(
        "expected_answer"
    )

    question_intent = output.get(
        "question_intent"
    )

    # -----------------------------------------------------
    # GENERATE_QA
    # -----------------------------------------------------

    if generation_decision == "GENERATE_QA":

        if (
            not isinstance(
                question,
                str,
            )
            or not question.strip()
        ):
            errors.append(
                "GENERATE_QA_QUESTION_INVALID"
            )

        else:

            normalized_question = (
                " ".join(
                    question.split()
                )
            )

            if any(
                phrase
                in normalized_question
                for phrase in (
                    GOLDEN_V2_FORBIDDEN_QUESTION_PHRASES
                )
            ):
                errors.append(
                    "QUESTION_HIDDEN_CONTEXT_DEPENDENCY"
                )

        if (
            not isinstance(
                expected_answer,
                str,
            )
            or not expected_answer.strip()
        ):
            errors.append(
                "GENERATE_QA_ANSWER_INVALID"
            )

        if (
            question_intent
            not in GOLDEN_V2_ALLOWED_QUESTION_INTENTS
        ):
            errors.append(
                "GENERATE_QA_INTENT_INVALID"
            )

    # -----------------------------------------------------
    # UNSUPPORTED_CLAIM
    # -----------------------------------------------------

    elif (
        generation_decision
        == "UNSUPPORTED_CLAIM"
    ):

        if question is not None:
            errors.append(
                "UNSUPPORTED_QUESTION_MUST_BE_NULL"
            )

        if expected_answer is not None:
            errors.append(
                "UNSUPPORTED_ANSWER_MUST_BE_NULL"
            )

        if question_intent is not None:
            errors.append(
                "UNSUPPORTED_INTENT_MUST_BE_NULL"
            )

    return errors


# =========================================================
# V2 PROTOCOL MANIFEST
# =========================================================

cell30_protocol_manifest_v2 = {
    "generation_protocol_version": (
        GOLDEN_V2_GENERATION_PROTOCOL_VERSION_V2
    ),

    "prompt_version": (
        GOLDEN_V2_GENERATION_PROMPT_VERSION_V2
    ),

    "output_schema_version": (
        GOLDEN_V2_GENERATION_OUTPUT_SCHEMA_VERSION_V2
    ),

    "record_contract_sha256": (
        cell25_record_contract_sha256
    ),

    "generation_decisions": list(
        GOLDEN_V2_GENERATION_DECISIONS_V2
    ),

    "allowed_question_intents": list(
        GOLDEN_V2_ALLOWED_QUESTION_INTENTS
    ),

    "claim_target_enforcement": True,

    "unsupported_claim_escape_hatch": True,

    "system_prompt": (
        GOLDEN_V2_GENERATION_SYSTEM_PROMPT_V2
    ),

    "output_contract": (
        cell30_generation_output_contract_v2
    ),
}


cell30_protocol_canonical_json_v2 = (
    json.dumps(
        cell30_protocol_manifest_v2,
        ensure_ascii=False,
        sort_keys=True,
        separators=(
            ",",
            ":",
        ),
    )
)


cell30_protocol_sha256_v2 = (
    hashlib.sha256(
        cell30_protocol_canonical_json_v2.encode(
            "utf-8"
        )
    ).hexdigest()
)


# =========================================================
# REQUEST FINGERPRINT
# =========================================================

def cell30_request_fingerprint_v2(
    skeleton,
):

    fingerprint_payload = {
        "record_contract_sha256": (
            cell25_record_contract_sha256
        ),

        "generation_protocol_sha256": (
            cell30_protocol_sha256_v2
        ),

        "candidate_id": (
            skeleton[
                "candidate_id"
            ]
        ),

        "user_prompt": (
            cell30_build_generation_user_prompt_v2(
                skeleton
            )
        ),
    }

    canonical = json.dumps(
        fingerprint_payload,
        ensure_ascii=False,
        sort_keys=True,
        separators=(
            ",",
            ":",
        ),
    )

    return hashlib.sha256(
        canonical.encode(
            "utf-8"
        )
    ).hexdigest()


# =========================================================
# BUILD ALL 3813 V2 REQUEST DESCRIPTORS
# =========================================================

cell30_generation_requests_v2 = []

cell30_request_errors_v2 = []


for skeleton in (
    cell25_candidate_skeletons
):

    try:

        payload = (
            cell30_build_generation_payload_v2(
                skeleton
            )
        )

        user_prompt = (
            cell30_build_generation_user_prompt_v2(
                skeleton
            )
        )

        request_fingerprint = (
            cell30_request_fingerprint_v2(
                skeleton
            )
        )

        cell30_generation_requests_v2.append({
            "candidate_id": (
                skeleton[
                    "candidate_id"
                ]
            ),

            "source_seed_slot_id": (
                skeleton[
                    "source_seed_slot_id"
                ]
            ),

            "claim_type": (
                skeleton[
                    "claim_type"
                ]
            ),

            "source_types": list(
                skeleton[
                    "source_types"
                ]
            ),

            "evidence_count": len(
                skeleton[
                    "evidence_ids"
                ]
            ),

            "payload": (
                payload
            ),

            "user_prompt": (
                user_prompt
            ),

            "request_fingerprint": (
                request_fingerprint
            ),
        })

    except Exception as exc:

        cell30_request_errors_v2.append({
            "candidate_id": (
                skeleton.get(
                    "candidate_id"
                )
            ),

            "error": (
                repr(
                    exc
                )
            ),
        })


# =========================================================
# REUSE EXACT SAME 12 CANARY IDS
# =========================================================

cell30_original_canary_ids = [
    request[
        "candidate_id"
    ]
    for request in (
        cell26_canary_requests
    )
]


cell30_request_v2_by_candidate_id = {
    request[
        "candidate_id"
    ]: request
    for request in (
        cell30_generation_requests_v2
    )
}


cell30_canary_requests_v2 = [
    cell30_request_v2_by_candidate_id[
        candidate_id
    ]
    for candidate_id in (
        cell30_original_canary_ids
    )
]


# =========================================================
# EXPECTED CANARY BEHAVIOR FROM CELL 29
#
# These are audit expectations, NOT forced model outputs.
# =========================================================

cell30_canary_expectations = {}


for record in (
    cell29_adjudicated_records
):

    old_decision = record[
        "decision"
    ]

    if (
        old_decision
        == "HOLD_UPSTREAM_EVIDENCE"
    ):

        expected_v2_behavior = (
            "EXPECTED_UNSUPPORTED_CLAIM"
        )

    elif (
        old_decision
        == "REJECT_GENERATION"
    ):

        expected_v2_behavior = (
            "EXPECTED_CLAIM_TARGETED_REGENERATION"
        )

    else:

        expected_v2_behavior = (
            "EXPECTED_VALID_GENERATE_QA"
        )

    cell30_canary_expectations[
        record[
            "candidate_id"
        ]
    ] = {
        "v1_adjudication": (
            old_decision
        ),

        "expected_v2_behavior": (
            expected_v2_behavior
        ),
    }


# =========================================================
# VALIDATION
# =========================================================

cell30_request_ids_v2 = [
    request[
        "candidate_id"
    ]
    for request in (
        cell30_generation_requests_v2
    )
]


cell30_request_fingerprints_v2 = [
    request[
        "request_fingerprint"
    ]
    for request in (
        cell30_generation_requests_v2
    )
]


cell30_validation_checks = {
    "3813_v2_requests_built": (
        len(
            cell30_generation_requests_v2
        )
        == 3813
    ),

    "zero_v2_request_errors": (
        len(
            cell30_request_errors_v2
        )
        == 0
    ),

    "v2_candidate_ids_unique": (
        len(
            set(
                cell30_request_ids_v2
            )
        )
        == 3813
    ),

    "v2_request_fingerprints_unique": (
        len(
            set(
                cell30_request_fingerprints_v2
            )
        )
        == 3813
    ),

    "exact_same_12_canary_ids_reused": (
        [
            request[
                "candidate_id"
            ]
            for request in (
                cell30_canary_requests_v2
            )
        ]
        == cell30_original_canary_ids
    ),

    "all_12_have_v2_expectations": (
        len(
            cell30_canary_expectations
        )
        == 12
    ),

    "two_upstream_holds_expect_unsupported": (
        sum(
            expectation[
                "expected_v2_behavior"
            ]
            == "EXPECTED_UNSUPPORTED_CLAIM"
            for expectation in (
                cell30_canary_expectations.values()
            )
        )
        == 2
    ),

    "one_generation_failure_requires_regeneration": (
        sum(
            expectation[
                "expected_v2_behavior"
            ]
            == "EXPECTED_CLAIM_TARGETED_REGENERATION"
            for expectation in (
                cell30_canary_expectations.values()
            )
        )
        == 1
    ),

    "protocol_v2_fingerprint_created": (
        len(
            cell30_protocol_sha256_v2
        )
        == 64
    ),
}


cell30_status = (
    "PASS"
    if all(
        cell30_validation_checks.values()
    )
    else "FAIL"
)


cell30_next_readiness = (
    "READY_FOR_GENERATION_V2_CANARY_EXECUTION"
    if cell30_status == "PASS"
    else "BLOCKED_GENERATION_V2_PROTOCOL_REPAIR_REQUIRED"
)


# =========================================================
# OUTPUT
# =========================================================

print(
    "=== GOLDEN V2 GENERATION PROTOCOL V2 DESIGN ==="
)

print()

print(
    "Protocol version:",
    GOLDEN_V2_GENERATION_PROTOCOL_VERSION_V2
)

print(
    "Prompt version:",
    GOLDEN_V2_GENERATION_PROMPT_VERSION_V2
)

print(
    "Output schema:",
    GOLDEN_V2_GENERATION_OUTPUT_SCHEMA_VERSION_V2
)

print(
    "Protocol SHA-256:",
    cell30_protocol_sha256_v2
)

print()

print(
    "V2 requests prepared:",
    len(
        cell30_generation_requests_v2
    )
)

print(
    "Request errors:",
    len(
        cell30_request_errors_v2
    )
)

print(
    "Unique fingerprints:",
    len(
        set(
            cell30_request_fingerprints_v2
        )
    )
)

print()

print(
    "Generation decisions:",
    list(
        GOLDEN_V2_GENERATION_DECISIONS_V2
    )
)

print()

print(
    "Canary v2 population:",
    len(
        cell30_canary_requests_v2
    )
)

print()

print(
    "V2 CANARY EXPECTATIONS:"
)

for index, candidate_id in enumerate(
    cell30_original_canary_ids,
    start=1,
):

    expectation = (
        cell30_canary_expectations[
            candidate_id
        ]
    )

    print(
        f" {index:02d}.",
        candidate_id,
        "| v1:",
        expectation[
            "v1_adjudication"
        ],
        "| v2 expectation:",
        expectation[
            "expected_v2_behavior"
        ],
    )


print()

print(
    "Validation checks:"
)

for (
    check_name,
    check_value,
) in cell30_validation_checks.items():

    print(
        " -",
        check_name,
        ":",
        (
            "PASS"
            if check_value
            else "FAIL"
        ),
    )

print()

print(
    "API calls performed:",
    0
)

print(
    "Files modified:",
    0
)

print(
    "Questions generated:",
    0
)

print(
    "Full Generation started:",
    False
)

print(
    "Next readiness:",
    cell30_next_readiness
)

print(
    "GOLDEN V2 GENERATION PROTOCOL V2 DESIGN:",
    cell30_status
)

assert cell30_status == "PASS", (
    "Golden v2 Generation Protocol v2 design failed."
)
````

به‌عنوان مرز اطلاعات مجاز آماده شده است. اکنون همان ۱۲ Candidate مرحله
Canary قبلی، بدون تغییر Population، دوباره با Protocol v2 اجرا می‌شوند.

برای موارد سالم، تشخیص UNSUPPORTED_CLAIM برای Claimهایی که Evidence کامل
آن‌ها را پشتیبانی نمی‌کند، و جلوگیری از ساخت سؤال درباره حقیقتی در Evidence
که خارج از Claim هدف است.

خواهد شد تا در مرحله بعد Semantic Adjudication نهایی Canary v2 انجام شود.
</p>

این مرحله فقط ۱۲ Canary را اجرا می‌کند. Full Generation برای ۳۸۱۳ Candidate

````python
# Code Cell 31 — Generation v2 Canary Checkpointed Execution
# =========================================================
# RERUN-SAFE / CHECKPOINTED / RESUMABLE
#
# Priority for every Canary:
#
#   1. Exact valid checkpoint
#   2. Migrate existing in-memory historical Cell 31 result
#   3. API only when neither exists
#
# Therefore:
# - Current kernel: preserve already-paid results
# - Restart Kernel -> Run All: load checkpoints, ZERO API
# - No duplicate paid request for an exact fingerprint
#
# IMPORTANT:
# Experimental FAIL is a valid historical outcome and is
# checkpointed too. We do NOT retry it merely because the
# experimental structural gate failed.
# =========================================================

from collections import Counter
from pathlib import Path
import hashlib
import json
import os
import re
import tempfile
import time


# =========================================================
# CONFIG
# =========================================================

CELL31_RUNNER_VERSION = (
    "golden_v2_generation_canary_v2_checkpoint_runner_v1"
)

GOLDEN_V2_CANARY_V2_MAX_TOKENS = 256
GOLDEN_V2_CANARY_V2_TEMPERATURE = 0.7
GOLDEN_V2_CANARY_V2_TOP_P = 0.8

GOLDEN_V2_CANARY_V2_MAX_ATTEMPTS = 2
GOLDEN_V2_CANARY_V2_RETRY_SECONDS = 2.0
GOLDEN_V2_CANARY_V2_REQUEST_SPACING_SECONDS = 2.7

# API is allowed ONLY for genuinely missing exact checkpoints.
CELL31_ALLOW_API_FOR_MISSING = False


# =========================================================
# REQUIRED STATE
# =========================================================

required_cell31_names = [
    "project_root",
    "semantic_review_client",
    "semantic_review_model",
    "cell30_canary_requests_v2",
    "cell30_canary_expectations",
    "cell30_build_generation_user_prompt_v2",
    "cell30_validate_generation_output_v2",
    "GOLDEN_V2_GENERATION_SYSTEM_PROMPT_V2",
    "GOLDEN_V2_GENERATION_PROTOCOL_VERSION_V2",
    "GOLDEN_V2_GENERATION_PROMPT_VERSION_V2",
    "GOLDEN_V2_GENERATION_OUTPUT_SCHEMA_VERSION_V2",
    "cell30_protocol_sha256_v2",
    "cell25_candidate_skeletons",
]

missing_cell31_names = [
    name
    for name in required_cell31_names
    if name not in globals()
]

assert not missing_cell31_names, (
    "Missing required Cell 31 state: "
    + ", ".join(missing_cell31_names)
)

assert len(cell30_canary_requests_v2) == 12


# =========================================================
# CAPTURE HISTORICAL IN-MEMORY RESULTS BEFORE OVERWRITE
# =========================================================

_cell31_existing_results_before_repair = globals().get(
    "cell31_canary_results_v2"
)

if (
    isinstance(
        _cell31_existing_results_before_repair,
        list,
    )
    and len(
        _cell31_existing_results_before_repair
    )
    == 12
):

    cell31_historical_results_available = True

    cell31_historical_results_by_candidate = {
        row.get("candidate_id"): row
        for row in (
            _cell31_existing_results_before_repair
        )
        if isinstance(row, dict)
        and row.get("candidate_id")
    }

else:

    cell31_historical_results_available = False
    cell31_historical_results_by_candidate = {}


# =========================================================
# CHECKPOINT DIRECTORY
# =========================================================

cell31_checkpoint_dir = (
    Path(project_root)
    / "artifacts"
    / "phase2"
    / "golden_dataset_v2"
    / "work"
    / "generation_canary_v2"
)

cell31_checkpoint_dir.mkdir(
    parents=True,
    exist_ok=True,
)


# =========================================================
# JSON / HASH UTILITIES
# =========================================================

def cell31_canonical_json(value):

    return json.dumps(
        value,
        ensure_ascii=False,
        sort_keys=True,
        separators=(",", ":"),
    )


def cell31_sha256_text(text):

    return hashlib.sha256(
        text.encode("utf-8")
    ).hexdigest()


def cell31_checkpoint_path(
    candidate_id,
):

    safe_id = re.sub(
        r"[^A-Za-z0-9_.-]+",
        "_",
        candidate_id,
    )

    return (
        cell31_checkpoint_dir
        / f"{safe_id}.json"
    )


def cell31_atomic_write_json(
    path,
    payload,
):

    path = Path(path)

    path.parent.mkdir(
        parents=True,
        exist_ok=True,
    )

    canonical = (
        json.dumps(
            payload,
            ensure_ascii=False,
            sort_keys=True,
            indent=2,
        )
        + "\n"
    )

    fd, temp_name = tempfile.mkstemp(
        prefix=path.name + ".",
        suffix=".tmp",
        dir=str(path.parent),
    )

    try:

        with os.fdopen(
            fd,
            "w",
            encoding="utf-8",
        ) as handle:

            handle.write(
                canonical
            )

            handle.flush()

            os.fsync(
                handle.fileno()
            )

        os.replace(
            temp_name,
            path,
        )

    except Exception:

        try:
            os.unlink(
                temp_name
            )
        except FileNotFoundError:
            pass

        raise


# =========================================================
# RESPONSE UTILITIES
# =========================================================

def cell31_clean_model_json(
    content,
):

    if not isinstance(
        content,
        str,
    ):
        return ""

    clean = content.strip()

    if clean.startswith("```"):

        clean = re.sub(
            r"^```(?:json)?\s*",
            "",
            clean,
            flags=re.IGNORECASE,
        )

        clean = re.sub(
            r"\s*```$",
            "",
            clean,
        )

    return clean.strip()


def cell31_is_rate_limit_exception(
    exc,
):

    error_text = str(exc).lower()

    error_type = (
        type(exc).__name__.lower()
    )

    return (
        "ratelimit" in error_type
        or "rate_limit" in error_type
        or "429" in error_text
        or "too many requests" in error_text
    )


# =========================================================
# COST RATE REUSE
# =========================================================

cell31_has_cost_rates = (
    "INPUT_TOMAN_PER_TOKEN" in globals()
    and
    "OUTPUT_TOMAN_PER_TOKEN" in globals()
)


def cell31_estimated_cost_toman(
    input_tokens,
    output_tokens,
):

    if not cell31_has_cost_rates:
        return None

    return (
        input_tokens
        * INPUT_TOMAN_PER_TOKEN
        +
        output_tokens
        * OUTPUT_TOMAN_PER_TOKEN
    )


# =========================================================
# SKELETON LOOKUP
# =========================================================

cell31_skeleton_by_candidate_id = {
    record["candidate_id"]: record
    for record in (
        cell25_candidate_skeletons
    )
}

assert len(
    cell31_skeleton_by_candidate_id
) == 3813


# =========================================================
# CHECKPOINT CONTRACT
# =========================================================

def cell31_build_checkpoint(
    *,
    request,
    result,
    provenance_mode,
):

    candidate_id = request[
        "candidate_id"
    ]

    checkpoint = {
        "checkpoint_version": (
            CELL31_RUNNER_VERSION
        ),

        "candidate_id": (
            candidate_id
        ),

        "request_fingerprint": (
            request[
                "request_fingerprint"
            ]
        ),

        "protocol_sha256": (
            cell30_protocol_sha256_v2
        ),

        "generation_protocol_version": (
            GOLDEN_V2_GENERATION_PROTOCOL_VERSION_V2
        ),

        "prompt_version": (
            GOLDEN_V2_GENERATION_PROMPT_VERSION_V2
        ),

        "output_schema_version": (
            GOLDEN_V2_GENERATION_OUTPUT_SCHEMA_VERSION_V2
        ),

        "model": (
            semantic_review_model
        ),

        "provenance_mode": (
            provenance_mode
        ),

        "historical_experimental_result": True,

        "result": (
            result
        ),
    }

    checkpoint[
        "checkpoint_payload_sha256"
    ] = cell31_sha256_text(
        cell31_canonical_json(
            checkpoint
        )
    )

    return checkpoint


def cell31_validate_checkpoint(
    checkpoint,
    request,
):

    errors = []

    if not isinstance(
        checkpoint,
        dict,
    ):
        return [
            "CHECKPOINT_NOT_OBJECT"
        ]

    stored_sha = checkpoint.get(
        "checkpoint_payload_sha256"
    )

    payload_without_sha = {
        key: value
        for key, value in checkpoint.items()
        if key
        != "checkpoint_payload_sha256"
    }

    recalculated_sha = (
        cell31_sha256_text(
            cell31_canonical_json(
                payload_without_sha
            )
        )
    )

    if (
        not isinstance(
            stored_sha,
            str,
        )
        or stored_sha
        != recalculated_sha
    ):
        errors.append(
            "CHECKPOINT_HASH_MISMATCH"
        )

    if (
        checkpoint.get(
            "candidate_id"
        )
        != request[
            "candidate_id"
        ]
    ):
        errors.append(
            "CHECKPOINT_CANDIDATE_MISMATCH"
        )

    if (
        checkpoint.get(
            "request_fingerprint"
        )
        != request[
            "request_fingerprint"
        ]
    ):
        errors.append(
            "CHECKPOINT_REQUEST_FINGERPRINT_MISMATCH"
        )

    if (
        checkpoint.get(
            "protocol_sha256"
        )
        != cell30_protocol_sha256_v2
    ):
        errors.append(
            "CHECKPOINT_PROTOCOL_MISMATCH"
        )

    if (
        checkpoint.get(
            "generation_protocol_version"
        )
        != GOLDEN_V2_GENERATION_PROTOCOL_VERSION_V2
    ):
        errors.append(
            "CHECKPOINT_GENERATION_PROTOCOL_MISMATCH"
        )

    if (
        checkpoint.get(
            "prompt_version"
        )
        != GOLDEN_V2_GENERATION_PROMPT_VERSION_V2
    ):
        errors.append(
            "CHECKPOINT_PROMPT_MISMATCH"
        )

    if (
        checkpoint.get(
            "output_schema_version"
        )
        != GOLDEN_V2_GENERATION_OUTPUT_SCHEMA_VERSION_V2
    ):
        errors.append(
            "CHECKPOINT_OUTPUT_SCHEMA_MISMATCH"
        )

    if (
        checkpoint.get(
            "model"
        )
        != semantic_review_model
    ):
        errors.append(
            "CHECKPOINT_MODEL_MISMATCH"
        )

    result = checkpoint.get(
        "result"
    )

    if not isinstance(
        result,
        dict,
    ):
        errors.append(
            "CHECKPOINT_RESULT_INVALID"
        )

    elif (
        result.get(
            "candidate_id"
        )
        != request[
            "candidate_id"
        ]
    ):
        errors.append(
            "CHECKPOINT_RESULT_CANDIDATE_MISMATCH"
        )

    return errors


def cell31_load_exact_checkpoint(
    request,
):

    path = cell31_checkpoint_path(
        request[
            "candidate_id"
        ]
    )

    if not path.exists():
        return (
            None,
            "MISSING",
            [],
        )

    try:

        with path.open(
            "r",
            encoding="utf-8",
        ) as handle:

            checkpoint = json.load(
                handle
            )

    except Exception as exc:

        return (
            None,
            "UNREADABLE",
            [
                (
                    f"{type(exc).__name__}: "
                    f"{exc}"
                )
            ],
        )

    errors = (
        cell31_validate_checkpoint(
            checkpoint,
            request,
        )
    )

    if errors:

        return (
            checkpoint,
            "INVALID",
            errors,
        )

    return (
        checkpoint,
        "VALID",
        [],
    )


# =========================================================
# HISTORICAL RESULT COMPATIBILITY
# =========================================================

def cell31_validate_historical_result(
    result,
    request,
):

    errors = []

    if not isinstance(
        result,
        dict,
    ):
        return [
            "HISTORICAL_RESULT_NOT_OBJECT"
        ]

    if (
        result.get(
            "candidate_id"
        )
        != request[
            "candidate_id"
        ]
    ):
        errors.append(
            "HISTORICAL_CANDIDATE_MISMATCH"
        )

    historical_fingerprint = (
        result.get(
            "request_fingerprint"
        )
    )

    if (
        historical_fingerprint
        is not None
        and historical_fingerprint
        != request[
            "request_fingerprint"
        ]
    ):
        errors.append(
            "HISTORICAL_REQUEST_FINGERPRINT_MISMATCH"
        )

    historical_protocol = (
        result.get(
            "generation_protocol_version"
        )
    )

    if (
        historical_protocol
        is not None
        and historical_protocol
        != GOLDEN_V2_GENERATION_PROTOCOL_VERSION_V2
    ):
        errors.append(
            "HISTORICAL_PROTOCOL_MISMATCH"
        )

    historical_prompt = (
        result.get(
            "prompt_version"
        )
    )

    if (
        historical_prompt
        is not None
        and historical_prompt
        != GOLDEN_V2_GENERATION_PROMPT_VERSION_V2
    ):
        errors.append(
            "HISTORICAL_PROMPT_MISMATCH"
        )

    historical_model = (
        result.get(
            "model"
        )
    )

    if (
        historical_model
        is not None
        and historical_model
        != semantic_review_model
    ):
        errors.append(
            "HISTORICAL_MODEL_MISMATCH"
        )

    return errors


# =========================================================
# API EXECUTION FOR GENUINELY MISSING ONLY
# =========================================================

def cell31_execute_missing_request(
    request,
):

    candidate_id = request[
        "candidate_id"
    ]

    skeleton = (
        cell31_skeleton_by_candidate_id[
            candidate_id
        ]
    )

    user_prompt = (
        cell30_build_generation_user_prompt_v2(
            skeleton
        )
    )

    expectation = (
        cell30_canary_expectations[
            candidate_id
        ][
            "expected_v2_behavior"
        ]
    )

    attempt_history = []

    for attempt_number in range(
        1,
        GOLDEN_V2_CANARY_V2_MAX_ATTEMPTS + 1,
    ):

        started_at = time.time()

        try:

            response = (
                semantic_review_client
                .chat
                .completions
                .create(
                    model=(
                        semantic_review_model
                    ),

                    messages=[
                        {
                            "role": "system",
                            "content": (
                                GOLDEN_V2_GENERATION_SYSTEM_PROMPT_V2
                            ),
                        },
                        {
                            "role": "user",
                            "content": (
                                user_prompt
                            ),
                        },
                    ],

                    temperature=(
                        GOLDEN_V2_CANARY_V2_TEMPERATURE
                    ),

                    top_p=(
                        GOLDEN_V2_CANARY_V2_TOP_P
                    ),

                    max_tokens=(
                        GOLDEN_V2_CANARY_V2_MAX_TOKENS
                    ),

                    extra_body={
                        "enable_thinking": False,
                    },
                )
            )

        except Exception as exc:

            attempt_history.append({
                "attempt": (
                    attempt_number
                ),
                "status": (
                    "API_ERROR"
                ),
                "error": (
                    f"{type(exc).__name__}: {exc}"
                ),
            })

            if (
                attempt_number
                < GOLDEN_V2_CANARY_V2_MAX_ATTEMPTS
            ):

                wait_seconds = (
                    70.0
                    if cell31_is_rate_limit_exception(
                        exc
                    )
                    else (
                        GOLDEN_V2_CANARY_V2_RETRY_SECONDS
                        * attempt_number
                    )
                )

                time.sleep(
                    wait_seconds
                )

                continue

            return {
                "candidate_id": (
                    candidate_id
                ),
                "execution_status": (
                    "API_FAILED"
                ),
                "expected_v2_behavior": (
                    expectation
                ),
                "attempt_history": (
                    attempt_history
                ),
            }

        elapsed_seconds = (
            time.time()
            - started_at
        )

        choice = response.choices[0]
        message = choice.message

        finish_reason = getattr(
            choice,
            "finish_reason",
            None,
        )

        raw_content = getattr(
            message,
            "content",
            None,
        )

        usage = getattr(
            response,
            "usage",
            None,
        )

        input_tokens = int(
            getattr(
                usage,
                "prompt_tokens",
                0,
            )
            or 0
        )

        output_tokens = int(
            getattr(
                usage,
                "completion_tokens",
                0,
            )
            or 0
        )

        total_tokens = int(
            getattr(
                usage,
                "total_tokens",
                0,
            )
            or (
                input_tokens
                + output_tokens
            )
        )

        reasoning_tokens = 0

        completion_details = getattr(
            usage,
            "completion_tokens_details",
            None,
        )

        if completion_details is not None:

            reasoning_tokens = int(
                getattr(
                    completion_details,
                    "reasoning_tokens",
                    0,
                )
                or 0
            )

        clean_content = (
            cell31_clean_model_json(
                raw_content
            )
        )

        parsed_json = None
        parse_error = None

        try:

            parsed_json = json.loads(
                clean_content
            )

        except Exception as exc:

            parse_error = (
                f"{type(exc).__name__}: {exc}"
            )

        if parsed_json is None:

            structural_errors = [
                "JSON_PARSE_FAILED"
            ]

        else:

            structural_errors = (
                cell30_validate_generation_output_v2(
                    parsed_json,
                    candidate_id,
                )
            )

        result = {
            "candidate_id": (
                candidate_id
            ),

            "source_seed_slot_id": (
                request[
                    "source_seed_slot_id"
                ]
            ),

            "claim_type": (
                request[
                    "claim_type"
                ]
            ),

            "source_types": (
                request[
                    "source_types"
                ]
            ),

            "evidence_count": (
                request[
                    "evidence_count"
                ]
            ),

            "request_fingerprint": (
                request[
                    "request_fingerprint"
                ]
            ),

            "expected_v2_behavior": (
                expectation
            ),

            "model": (
                semantic_review_model
            ),

            "generation_protocol_version": (
                GOLDEN_V2_GENERATION_PROTOCOL_VERSION_V2
            ),

            "prompt_version": (
                GOLDEN_V2_GENERATION_PROMPT_VERSION_V2
            ),

            "output_schema_version": (
                GOLDEN_V2_GENERATION_OUTPUT_SCHEMA_VERSION_V2
            ),

            "attempt_number": (
                attempt_number
            ),

            "finish_reason": (
                finish_reason
            ),

            "elapsed_seconds": round(
                elapsed_seconds,
                4,
            ),

            "usage": {
                "input_tokens": (
                    input_tokens
                ),
                "output_tokens": (
                    output_tokens
                ),
                "total_tokens": (
                    total_tokens
                ),
                "reasoning_tokens": (
                    reasoning_tokens
                ),
            },

            "estimated_cost_toman": (
                cell31_estimated_cost_toman(
                    input_tokens,
                    output_tokens,
                )
            ),

            "raw_content": (
                raw_content
            ),

            "parsed_json": (
                parsed_json
            ),

            "parse_error": (
                parse_error
            ),

            "structural_errors": (
                structural_errors
            ),

            "structural_status": (
                "PASS"
                if not structural_errors
                else "FAIL"
            ),

            "attempt_history": list(
                attempt_history
            ),
        }

        # IMPORTANT:
        # Once a paid response exists, preserve it.
        # Experimental schema FAIL is still a completed
        # historical observation and must not trigger
        # repeated paid execution on future Run All.
        return result

    raise RuntimeError(
        "Unexpected Cell 31 execution path."
    )


# =========================================================
# PREFLIGHT
# =========================================================

print(
    "=== GOLDEN V2 GENERATION V2 CANARY "
    "CHECKPOINT PREFLIGHT ==="
)

print()

print(
    "Canary population:",
    len(
        cell30_canary_requests_v2
    ),
)

print(
    "Model:",
    semantic_review_model,
)

print(
    "Protocol:",
    GOLDEN_V2_GENERATION_PROTOCOL_VERSION_V2,
)

print(
    "Protocol SHA-256:",
    cell30_protocol_sha256_v2,
)

print(
    "Checkpoint directory:",
    cell31_checkpoint_dir.relative_to(
        Path(project_root)
    ),
)

print(
    "Historical in-memory result set available:",
    cell31_historical_results_available,
)

print(
    "API allowed for genuinely missing exact results:",
    CELL31_ALLOW_API_FOR_MISSING,
)

print()


# =========================================================
# RESOLVE EACH CANARY
# =========================================================

cell31_canary_results_v2 = []

cell31_resolution_modes = []

cell31_checkpoint_validation_issues = []

cell31_api_calls_this_execution = 0

cell31_checkpoints_created_this_execution = 0


for index, request in enumerate(
    cell30_canary_requests_v2,
    start=1,
):

    candidate_id = request[
        "candidate_id"
    ]

    checkpoint, checkpoint_status, checkpoint_errors = (
        cell31_load_exact_checkpoint(
            request
        )
    )

    # -----------------------------------------------------
    # 1. EXACT VALID CHECKPOINT
    # -----------------------------------------------------

    if checkpoint_status == "VALID":

        result = checkpoint[
            "result"
        ]

        resolution_mode = (
            "EXACT_CHECKPOINT"
        )

    # -----------------------------------------------------
    # 2. MIGRATE HISTORICAL IN-MEMORY RESULT
    # -----------------------------------------------------

    elif (
        candidate_id
        in cell31_historical_results_by_candidate
    ):

        historical_result = (
            cell31_historical_results_by_candidate[
                candidate_id
            ]
        )

        historical_errors = (
            cell31_validate_historical_result(
                historical_result,
                request,
            )
        )

        if historical_errors:

            raise AssertionError(
                "Historical Cell 31 result is not "
                "compatible with exact v2 request for "
                f"{candidate_id}: "
                f"{historical_errors}"
            )

        checkpoint = (
            cell31_build_checkpoint(
                request=request,
                result=historical_result,
                provenance_mode=(
                    "MIGRATED_FROM_IN_MEMORY_"
                    "HISTORICAL_EXECUTION"
                ),
            )
        )

        checkpoint_path = (
            cell31_checkpoint_path(
                candidate_id
            )
        )

        cell31_atomic_write_json(
            checkpoint_path,
            checkpoint,
        )

        # Immediate read-back validation
        (
            verified_checkpoint,
            verified_status,
            verified_errors,
        ) = cell31_load_exact_checkpoint(
            request
        )

        assert (
            verified_status
            == "VALID"
        ), (
            "Checkpoint migration verification failed "
            f"for {candidate_id}: "
            f"{verified_errors}"
        )

        result = (
            verified_checkpoint[
                "result"
            ]
        )

        resolution_mode = (
            "MIGRATED_HISTORICAL_RESULT"
        )

        cell31_checkpoints_created_this_execution += 1

    # -----------------------------------------------------
    # 3. API ONLY IF NO EXACT RESULT EXISTS
    # -----------------------------------------------------

    else:

        if checkpoint_status in {
            "INVALID",
            "UNREADABLE",
        }:

            cell31_checkpoint_validation_issues.append({
                "candidate_id": (
                    candidate_id
                ),
                "checkpoint_status": (
                    checkpoint_status
                ),
                "errors": (
                    checkpoint_errors
                ),
            })

            raise AssertionError(
                "A checkpoint exists but is invalid. "
                "Refusing automatic paid replacement for "
                f"{candidate_id}. "
                f"Status={checkpoint_status}; "
                f"errors={checkpoint_errors}"
            )

        if not CELL31_ALLOW_API_FOR_MISSING:

            raise AssertionError(
                "Exact checkpoint missing and API execution "
                "is disabled for "
                f"{candidate_id}."
            )

        if (
            cell31_api_calls_this_execution
            > 0
        ):

            time.sleep(
                GOLDEN_V2_CANARY_V2_REQUEST_SPACING_SECONDS
            )

        result = (
            cell31_execute_missing_request(
                request
            )
        )

        cell31_api_calls_this_execution += 1

        checkpoint = (
            cell31_build_checkpoint(
                request=request,
                result=result,
                provenance_mode=(
                    "NEW_PAID_EXECUTION"
                ),
            )
        )

        checkpoint_path = (
            cell31_checkpoint_path(
                candidate_id
            )
        )

        cell31_atomic_write_json(
            checkpoint_path,
            checkpoint,
        )

        (
            verified_checkpoint,
            verified_status,
            verified_errors,
        ) = cell31_load_exact_checkpoint(
            request
        )

        assert (
            verified_status
            == "VALID"
        ), (
            "New checkpoint verification failed "
            f"for {candidate_id}: "
            f"{verified_errors}"
        )

        result = (
            verified_checkpoint[
                "result"
            ]
        )

        resolution_mode = (
            "NEW_PAID_EXECUTION"
        )

        cell31_checkpoints_created_this_execution += 1

    cell31_canary_results_v2.append(
        result
    )

    cell31_resolution_modes.append({
        "candidate_id": (
            candidate_id
        ),
        "resolution_mode": (
            resolution_mode
        ),
    })

    print(
        f"[{index:02d}/12]",
        candidate_id,
        "|",
        resolution_mode,
        "| structural:",
        result.get(
            "structural_status",
            result.get(
                "execution_status"
            ),
        ),
    )


# =========================================================
# FINAL DISK RESCAN
# =========================================================

cell31_final_valid_checkpoint_count = 0
cell31_final_invalid_checkpoint_count = 0
cell31_final_missing_checkpoint_count = 0

cell31_final_checkpoint_errors = []


for request in (
    cell30_canary_requests_v2
):

    (
        checkpoint,
        status,
        errors,
    ) = cell31_load_exact_checkpoint(
        request
    )

    if status == "VALID":

        cell31_final_valid_checkpoint_count += 1

    elif status == "MISSING":

        cell31_final_missing_checkpoint_count += 1

        cell31_final_checkpoint_errors.append({
            "candidate_id": (
                request[
                    "candidate_id"
                ]
            ),
            "status": status,
            "errors": errors,
        })

    else:

        cell31_final_invalid_checkpoint_count += 1

        cell31_final_checkpoint_errors.append({
            "candidate_id": (
                request[
                    "candidate_id"
                ]
            ),
            "status": status,
            "errors": errors,
        })


# =========================================================
# RECONSTRUCT EXPERIMENTAL SUMMARY
# =========================================================

cell31_structural_pass_results_v2 = [
    record
    for record in (
        cell31_canary_results_v2
    )
    if record.get(
        "structural_status"
    )
    == "PASS"
]

cell31_structural_pass_count_v2 = len(
    cell31_structural_pass_results_v2
)

cell31_failure_count_v2 = (
    12
    - cell31_structural_pass_count_v2
)


cell31_decision_distribution_v2 = Counter(
    record[
        "parsed_json"
    ][
        "generation_decision"
    ]
    for record in (
        cell31_structural_pass_results_v2
    )
    if isinstance(
        record.get(
            "parsed_json"
        ),
        dict,
    )
)


# =========================================================
# EXPECTATION DIAGNOSTICS
# =========================================================

cell31_expectation_diagnostics = []


for record in (
    cell31_structural_pass_results_v2
):

    candidate_id = record[
        "candidate_id"
    ]

    expected_behavior = (
        cell30_canary_expectations[
            candidate_id
        ][
            "expected_v2_behavior"
        ]
    )

    observed_decision = (
        record[
            "parsed_json"
        ][
            "generation_decision"
        ]
    )

    if (
        expected_behavior
        == "EXPECTED_UNSUPPORTED_CLAIM"
    ):

        expectation_match = (
            observed_decision
            == "UNSUPPORTED_CLAIM"
        )

    elif expected_behavior in {
        "EXPECTED_VALID_GENERATE_QA",
        "EXPECTED_CLAIM_TARGETED_REGENERATION",
    }:

        expectation_match = (
            observed_decision
            == "GENERATE_QA"
        )

    else:

        expectation_match = False

    cell31_expectation_diagnostics.append({
        "candidate_id": (
            candidate_id
        ),
        "expected_behavior": (
            expected_behavior
        ),
        "observed_decision": (
            observed_decision
        ),
        "decision_class_match": (
            expectation_match
        ),
    })


cell31_decision_class_match_count = sum(
    row[
        "decision_class_match"
    ]
    for row in (
        cell31_expectation_diagnostics
    )
)


# =========================================================
# TOKEN SUMMARY — HISTORICAL OBSERVATION
# =========================================================

cell31_total_input_tokens_v2 = sum(
    (
        record.get(
            "usage"
        )
        or {}
    ).get(
        "input_tokens",
        0,
    )
    for record in (
        cell31_canary_results_v2
    )
)

cell31_total_output_tokens_v2 = sum(
    (
        record.get(
            "usage"
        )
        or {}
    ).get(
        "output_tokens",
        0,
    )
    for record in (
        cell31_canary_results_v2
    )
)

cell31_total_tokens_v2 = sum(
    (
        record.get(
            "usage"
        )
        or {}
    ).get(
        "total_tokens",
        0,
    )
    for record in (
        cell31_canary_results_v2
    )
)

cell31_total_reasoning_tokens_v2 = sum(
    (
        record.get(
            "usage"
        )
        or {}
    ).get(
        "reasoning_tokens",
        0,
    )
    for record in (
        cell31_canary_results_v2
    )
)


# =========================================================
# TWO SEPARATE STATUSES
#
# Experimental semantic/schema outcome can remain FAIL.
# Runner health MUST PASS.
# =========================================================

cell31_experimental_status = (
    "PASS"
    if (
        cell31_structural_pass_count_v2
        == 12
    )
    else "FAIL"
)


cell31_runner_health_checks = {
    "exactly_12_results_reconstructed": (
        len(
            cell31_canary_results_v2
        )
        == 12
    ),

    "all_12_exact_checkpoints_exist": (
        cell31_final_valid_checkpoint_count
        == 12
    ),

    "zero_missing_checkpoints": (
        cell31_final_missing_checkpoint_count
        == 0
    ),

    "zero_invalid_checkpoints": (
        cell31_final_invalid_checkpoint_count
        == 0
    ),

    "zero_checkpoint_validation_issues": (
        len(
            cell31_checkpoint_validation_issues
        )
        == 0
    ),

    "full_generation_not_started": True,
}


cell31_runner_status = (
    "PASS"
    if all(
        cell31_runner_health_checks.values()
    )
    else "FAIL"
)


cell31_status = (
    cell31_experimental_status
)


cell31_next_readiness = (
    "READY_FOR_CANARY_V2_FAILURE_ANALYSIS"
    if cell31_runner_status == "PASS"
    else "BLOCKED_CELL31_RUNNER_REPAIR_REQUIRED"
)


# =========================================================
# OUTPUT
# =========================================================

print()

print(
    "=== GOLDEN V2 GENERATION V2 CANARY "
    "CHECKPOINT RESULT ==="
)

print()

print(
    "Results reconstructed:",
    len(
        cell31_canary_results_v2
    ),
)

print(
    "Exact valid checkpoints:",
    cell31_final_valid_checkpoint_count,
)

print(
    "Missing checkpoints:",
    cell31_final_missing_checkpoint_count,
)

print(
    "Invalid checkpoints:",
    cell31_final_invalid_checkpoint_count,
)

print()

print(
    "Resolution modes:",
    dict(
        Counter(
            row[
                "resolution_mode"
            ]
            for row in (
                cell31_resolution_modes
            )
        )
    ),
)

print(
    "Checkpoints created this execution:",
    cell31_checkpoints_created_this_execution,
)

print(
    "API calls this execution:",
    cell31_api_calls_this_execution,
)

print()

print(
    "Historical Structural PASS:",
    cell31_structural_pass_count_v2,
    "/ 12",
)

print(
    "Historical experimental outcome:",
    cell31_experimental_status,
)

print(
    "Decision distribution:",
    dict(
        cell31_decision_distribution_v2
    ),
)

print(
    "Decision-class expectation matches:",
    cell31_decision_class_match_count,
    "/ 12",
)

print()

print(
    "Historical input tokens:",
    cell31_total_input_tokens_v2,
)

print(
    "Historical output tokens:",
    cell31_total_output_tokens_v2,
)

print(
    "Historical total tokens:",
    cell31_total_tokens_v2,
)

print(
    "Historical reasoning tokens:",
    cell31_total_reasoning_tokens_v2,
)

print()

print(
    "Runner health checks:"
)

for (
    check_name,
    check_value,
) in (
    cell31_runner_health_checks.items()
):

    print(
        " -",
        check_name,
        ":",
        (
            "PASS"
            if check_value
            else "FAIL"
        ),
    )

print()

print(
    "Runner health:",
    cell31_runner_status,
)

print(
    "Experimental result:",
    cell31_experimental_status,
)

print(
    "Full Generation started:",
    False,
)

print(
    "Next readiness:",
    cell31_next_readiness,
)

print(
    "CELL 31 CHECKPOINTED RUNNER:",
    cell31_runner_status,
)

# Only runner/infrastructure failure may stop Run All.
# Experimental FAIL is preserved as data, not raised.
assert cell31_runner_status == "PASS", (
    "Cell 31 checkpointed runner health failed."
)
````

<p>
در این مرحله قرارداد نهایی تولید سؤال و پاسخ برای Golden Dataset v2 ساخته می‌شود.
اعتبار Claim و ارتباط Claim با Evidence قبلاً در NB12i اعتبارسنجی شده است؛ بنابراین این مرحله آن بررسی را تکرار نمی‌کند.
وظیفه مدل فقط تولید یک سؤال مستقل و دقیق و یک پاسخ مورد انتظار است که مستقیماً Claim هدف را پوشش دهند و از محتوای Evidence فراتر نروند.
در این مرحله هیچ فراخوانی API انجام نمی‌شود.
</p>

````python
# Code Cell 32 — Final QA Generation v3 Protocol

import json
import hashlib
from pathlib import Path

print("=== GOLDEN V2 FINAL QA GENERATION V3 PROTOCOL ===")

# ------------------------------------------------------------
# 1. Preconditions from completed notebook state
# ------------------------------------------------------------

required_names = [
    "cell25_candidate_skeletons",
]

missing_names = [name for name in required_names if name not in globals()]

if missing_names:
    raise RuntimeError(
        "Required upstream notebook state is missing: "
        + ", ".join(missing_names)
    )

records = cell25_candidate_skeletons

if not isinstance(records, list):
    raise TypeError("golden_v2_record_skeletons must be a list.")

if len(records) != 3813:
    raise RuntimeError(
        f"Expected 3813 accepted Golden v2 record skeletons, got {len(records)}."
    )

# ------------------------------------------------------------
# 2. Final generation architecture
# ------------------------------------------------------------

GENERATION_V3_VERSION = "golden_v2_question_answer_generation_v3"

GENERATION_V3_SYSTEM_PROMPT = """
You generate one evaluation question and one expected answer for a
pre-validated Golden Dataset candidate.

Important architecture rule:
The target Claim and its Evidence have already passed upstream semantic
validation. Do NOT re-adjudicate whether the Claim is supported.
Do NOT output ACCEPT, REJECT, UNSUPPORTED, or any support verdict.

Your only task is to generate a high-fidelity Question and Expected Answer.

Requirements:

1. The Question must directly target the supplied Claim.
2. The Question must not shift to a neighboring fact merely because that
   neighboring fact also appears in the Evidence.
3. A correct answer to the Question must substantially recover the meaning
   of the target Claim.
4. The Expected Answer must preserve the Claim's:
   - meaning
   - scope
   - modality
   - quantifiers
   - conditions
   - perspective
   when present.
5. The Expected Answer must not introduce unsupported external knowledge.
6. The Question must be standalone, clear, and unambiguous.
7. Do not mention "claim", "evidence", "dataset", "context", or the generation
   process in the Question.
8. Do not ask for information broader than the target Claim.
9. Do not make the Question artificially difficult.
10. Use the Evidence only as grounding/context for expressing the target
    Claim faithfully; do not replace the target Claim with another assertion
    from the Evidence.

Return JSON only with exactly these fields:

{
  "question": "...",
  "expected_answer": "...",
  "question_intent": "..."
}
""".strip()

GENERATION_V3_OUTPUT_SCHEMA = {
    "type": "object",
    "required": [
        "question",
        "expected_answer",
        "question_intent",
    ],
    "additionalProperties": False,
    "properties": {
        "question": {
            "type": "string",
            "minLength": 1,
        },
        "expected_answer": {
            "type": "string",
            "minLength": 1,
        },
        "question_intent": {
            "type": "string",
            "minLength": 1,
        },
    },
}

# ------------------------------------------------------------
# 3. Canonical protocol object
# ------------------------------------------------------------

generation_v3_protocol = {
    "protocol_version": GENERATION_V3_VERSION,
    "architecture": {
        "claim_evidence_revalidation": False,
        "support_verdict_generation": False,
        "material_assertion_extraction": False,
        "qa_generation_only": True,
    },
    "input_population": {
        "record_count": len(records),
        "population_role": "post_semantic_review_accepted_seed_population",
    },
    "generation_requirements": {
        "question_targets_claim": True,
        "answer_recovers_claim_meaning": True,
        "question_standalone": True,
        "question_unambiguous": True,
        "preserve_scope": True,
        "preserve_modality": True,
        "preserve_quantifiers": True,
        "preserve_conditions": True,
        "preserve_perspective": True,
        "no_external_knowledge": True,
        "no_neighbor_fact_drift": True,
    },
    "output_fields": [
        "question",
        "expected_answer",
        "question_intent",
    ],
    "system_prompt": GENERATION_V3_SYSTEM_PROMPT,
    "output_schema": GENERATION_V3_OUTPUT_SCHEMA,
}

protocol_bytes = json.dumps(
    generation_v3_protocol,
    ensure_ascii=False,
    sort_keys=True,
    separators=(",", ":"),
).encode("utf-8")

GENERATION_V3_PROTOCOL_SHA256 = hashlib.sha256(protocol_bytes).hexdigest()

generation_v3_protocol["protocol_sha256"] = GENERATION_V3_PROTOCOL_SHA256

# ------------------------------------------------------------
# 4. Zero-API design checks
# ------------------------------------------------------------

checks = {
    "accepted_population_is_3813": len(records) == 3813,
    "qa_generation_only": (
        generation_v3_protocol["architecture"]["qa_generation_only"] is True
    ),
    "no_claim_evidence_revalidation": (
        generation_v3_protocol["architecture"]["claim_evidence_revalidation"]
        is False
    ),
    "no_support_verdict_generation": (
        generation_v3_protocol["architecture"]["support_verdict_generation"]
        is False
    ),
    "no_material_assertion_branch": (
        generation_v3_protocol["architecture"]["material_assertion_extraction"]
        is False
    ),
    "three_output_fields_only": (
        generation_v3_protocol["output_fields"]
        == ["question", "expected_answer", "question_intent"]
    ),
    "protocol_hash_created": (
        isinstance(GENERATION_V3_PROTOCOL_SHA256, str)
        and len(GENERATION_V3_PROTOCOL_SHA256) == 64
    ),
}

failed_checks = [name for name, passed in checks.items() if not passed]

print(f"Protocol version: {GENERATION_V3_VERSION}")
print(f"Target population: {len(records)}")
print(f"Protocol SHA-256: {GENERATION_V3_PROTOCOL_SHA256}")
print("API calls: 0")
print()

for name, passed in checks.items():
    print(f"{name}: {'PASS' if passed else 'FAIL'}")

print()

if failed_checks:
    print("CELL 32 STATUS: FAIL")
    print("Failed checks:", failed_checks)
    raise RuntimeError("Generation v3 protocol design failed.")

print("CELL 32 STATUS: PASS")
print("Next readiness: READY_FOR_QA_GENERATION_V3_CANARY")
````

<div style="direction: rtl; white-space: normal; line-height: 1.7;">
<h2>مرحله ۳۳ — اجرای Canary تولید سؤال و پاسخ با Generation v3</h2>
<p>
در این مرحله Generation v3 روی همان ۱۲ نمونه Canary قبلی اجرا می‌شود تا کیفیت روش جدید مستقیماً با آزمایش‌های قبلی قابل مقایسه باشد.
هر درخواست با ترکیب دقیق Candidate، مدل، تنظیمات و Protocol v3 دارای Fingerprint مستقل است و نتیجه آن به‌صورت اتمیک Checkpoint می‌شود.
وجود Checkpoint معتبر از اجرای دوباره و هزینه تکراری جلوگیری می‌کند.
این مرحله فقط Canary است و تولید کامل ۳۸۱۳ رکورد را آغاز نمی‌کند.
</p>
</div>

````python
# Code Cell 33 — QA Generation v3 Canary Checkpointed Execution

from collections import Counter
from pathlib import Path
import hashlib
import json
import os
import re
import tempfile
import time


# =========================================================
# CONFIG
# =========================================================

CELL33_RUNNER_VERSION = (
    "golden_v2_generation_canary_v3_checkpoint_runner_v1"
)

CELL33_PROMPT_VERSION = (
    "golden_v2_qa_generation_prompt_v3"
)

CELL33_OUTPUT_SCHEMA_VERSION = (
    "golden_v2_qa_generation_output_v3"
)

CELL33_MAX_TOKENS = 256
CELL33_TEMPERATURE = 0.7
CELL33_TOP_P = 0.8

CELL33_MAX_ATTEMPTS = 2
CELL33_RETRY_SECONDS = 2.0
CELL33_REQUEST_SPACING_SECONDS = 2.7

# This is an intentional paid Canary run.
# Before the final Restart -> Run All health check,
# this flag MUST be changed to False.
CELL33_ALLOW_API_FOR_MISSING = False


# =========================================================
# REQUIRED STATE
# =========================================================

required_cell33_names = [
    "project_root",
    "semantic_review_client",
    "semantic_review_model",
    "cell25_candidate_skeletons",
    "cell26_canary_requests",
    "GENERATION_V3_VERSION",
    "GENERATION_V3_SYSTEM_PROMPT",
    "GENERATION_V3_PROTOCOL_SHA256",
]

missing_cell33_names = [
    name
    for name in required_cell33_names
    if name not in globals()
]

assert not missing_cell33_names, (
    "Missing required Cell 33 state: "
    + ", ".join(missing_cell33_names)
)

assert len(cell25_candidate_skeletons) == 3813
assert len(cell26_canary_requests) == 12

assert (
    semantic_review_model.lower()
    == "qwen3-30b-a3b".lower()
), (
    "Unexpected semantic_review_model: "
    f"{semantic_review_model}"
)


# =========================================================
# LOOKUPS
# =========================================================

cell33_skeleton_by_candidate_id = {
    record["candidate_id"]: record
    for record in cell25_candidate_skeletons
}

assert len(cell33_skeleton_by_candidate_id) == 3813

cell33_canary_candidate_ids = [
    request["candidate_id"]
    for request in cell26_canary_requests
]

assert len(cell33_canary_candidate_ids) == 12
assert len(set(cell33_canary_candidate_ids)) == 12

missing_canary_candidates = [
    candidate_id
    for candidate_id in cell33_canary_candidate_ids
    if candidate_id not in cell33_skeleton_by_candidate_id
]

assert not missing_canary_candidates, (
    "Canary candidates missing from Cell 25 skeletons: "
    + ", ".join(missing_canary_candidates)
)


# =========================================================
# CANONICAL JSON / HASH
# =========================================================

def cell33_canonical_json(value):

    return json.dumps(
        value,
        ensure_ascii=False,
        sort_keys=True,
        separators=(",", ":"),
    )


def cell33_sha256_text(text):

    return hashlib.sha256(
        text.encode("utf-8")
    ).hexdigest()


# =========================================================
# USER PAYLOAD / PROMPT
# =========================================================

def cell33_build_payload(
    skeleton,
):

    evidence = []

    for (
        evidence_id,
        evidence_quote,
        evidence_locator,
    ) in zip(
        skeleton["evidence_ids"],
        skeleton["evidence_quotes"],
        skeleton["evidence_locators"],
    ):

        evidence.append({
            "evidence_id": evidence_id,
            "quote": evidence_quote,
            "locator": evidence_locator,
        })

    return {
        "candidate_id": skeleton["candidate_id"],

        "target_claim": {
            "claim_id": skeleton["claim_id"],
            "text_fa": skeleton["claim_text_fa"],
            "claim_type": skeleton["claim_type"],
            "claim_types": skeleton["claim_types"],
            "authority_class": skeleton["authority_class"],
        },

        "evidence": evidence,

        "generation_instruction": (
            "Generate exactly one Persian standalone question "
            "and one Persian expected answer that directly "
            "evaluate the target Claim. Evidence is grounding "
            "context only. Do not re-adjudicate Claim support."
        ),

        "required_output_fields": [
            "question",
            "expected_answer",
            "question_intent",
        ],
    }


def cell33_build_user_prompt(
    skeleton,
):

    payload = cell33_build_payload(
        skeleton
    )

    return (
        "Generate the requested Golden Dataset evaluation "
        "question and expected answer from this JSON input.\n\n"
        + json.dumps(
            payload,
            ensure_ascii=False,
            sort_keys=True,
            indent=2,
        )
    )


# =========================================================
# OUTPUT VALIDATION
# =========================================================

def cell33_validate_output(
    parsed_json,
):

    errors = []

    if not isinstance(parsed_json, dict):
        return ["OUTPUT_NOT_OBJECT"]

    expected_fields = {
        "question",
        "expected_answer",
        "question_intent",
    }

    actual_fields = set(
        parsed_json.keys()
    )

    if actual_fields != expected_fields:
        errors.append(
            "OUTPUT_FIELDS_MISMATCH"
        )

    for field_name in expected_fields:

        value = parsed_json.get(
            field_name
        )

        if (
            not isinstance(value, str)
            or not value.strip()
        ):
            errors.append(
                f"INVALID_{field_name.upper()}"
            )

    question = parsed_json.get(
        "question"
    )

    if isinstance(question, str):

        forbidden_question_phrases = [
            "طبق متن بالا",
            "در متن بالا",
            "طبق شواهد",
            "بر اساس شواهد",
            "طبق claim",
            "طبق evidence",
            "در dataset",
        ]

        question_lower = (
            question.lower()
        )

        for phrase in forbidden_question_phrases:

            if phrase.lower() in question_lower:

                errors.append(
                    "QUESTION_CONTAINS_CONTEXT_REFERENCE"
                )

                break

    return sorted(
        set(errors)
    )


# =========================================================
# RESPONSE CLEANING
# =========================================================

def cell33_clean_model_json(
    content,
):

    if not isinstance(content, str):
        return ""

    clean = content.strip()

    if clean.startswith("```"):

        clean = re.sub(
            r"^```(?:json)?\s*",
            "",
            clean,
            flags=re.IGNORECASE,
        )

        clean = re.sub(
            r"\s*```$",
            "",
            clean,
        )

    return clean.strip()


def cell33_is_rate_limit_exception(
    exc,
):

    error_text = str(exc).lower()
    error_type = type(exc).__name__.lower()

    return (
        "ratelimit" in error_type
        or "rate_limit" in error_type
        or "429" in error_text
        or "too many requests" in error_text
    )


# =========================================================
# REQUEST CONTRACT
# =========================================================

def cell33_build_request(
    skeleton,
):

    candidate_id = skeleton[
        "candidate_id"
    ]

    user_prompt = (
        cell33_build_user_prompt(
            skeleton
        )
    )

    fingerprint_payload = {
        "runner_version": (
            CELL33_RUNNER_VERSION
        ),

        "candidate_id": (
            candidate_id
        ),

        "model": (
            semantic_review_model
        ),

        "protocol_version": (
            GENERATION_V3_VERSION
        ),

        "protocol_sha256": (
            GENERATION_V3_PROTOCOL_SHA256
        ),

        "prompt_version": (
            CELL33_PROMPT_VERSION
        ),

        "output_schema_version": (
            CELL33_OUTPUT_SCHEMA_VERSION
        ),

        "system_prompt": (
            GENERATION_V3_SYSTEM_PROMPT
        ),

        "user_prompt": (
            user_prompt
        ),

        "temperature": (
            CELL33_TEMPERATURE
        ),

        "top_p": (
            CELL33_TOP_P
        ),

        "max_tokens": (
            CELL33_MAX_TOKENS
        ),

        "enable_thinking": False,
    }

    request_fingerprint = (
        cell33_sha256_text(
            cell33_canonical_json(
                fingerprint_payload
            )
        )
    )

    return {
        "candidate_id": (
            candidate_id
        ),

        "source_seed_slot_id": (
            skeleton[
                "source_seed_slot_id"
            ]
        ),

        "claim_id": (
            skeleton[
                "claim_id"
            ]
        ),

        "claim_type": (
            skeleton[
                "claim_type"
            ]
        ),

        "evidence_count": len(
            skeleton[
                "evidence_ids"
            ]
        ),

        "user_prompt": (
            user_prompt
        ),

        "request_fingerprint": (
            request_fingerprint
        ),
    }


cell33_canary_requests = [
    cell33_build_request(
        cell33_skeleton_by_candidate_id[
            candidate_id
        ]
    )
    for candidate_id in (
        cell33_canary_candidate_ids
    )
]

assert len(cell33_canary_requests) == 12


# =========================================================
# CHECKPOINT DIRECTORY
# =========================================================

cell33_checkpoint_dir = (
    Path(project_root)
    / "artifacts"
    / "phase2"
    / "golden_dataset_v2"
    / "work"
    / "generation_canary_v3"
)

cell33_checkpoint_dir.mkdir(
    parents=True,
    exist_ok=True,
)


def cell33_checkpoint_path(
    candidate_id,
):

    safe_id = re.sub(
        r"[^A-Za-z0-9_.-]+",
        "_",
        candidate_id,
    )

    return (
        cell33_checkpoint_dir
        / f"{safe_id}.json"
    )


# =========================================================
# ATOMIC CHECKPOINT WRITE
# =========================================================

def cell33_atomic_write_json(
    path,
    payload,
):

    path = Path(path)

    path.parent.mkdir(
        parents=True,
        exist_ok=True,
    )

    content = (
        json.dumps(
            payload,
            ensure_ascii=False,
            sort_keys=True,
            indent=2,
        )
        + "\n"
    )

    fd, temp_name = tempfile.mkstemp(
        prefix=path.name + ".",
        suffix=".tmp",
        dir=str(path.parent),
    )

    try:

        with os.fdopen(
            fd,
            "w",
            encoding="utf-8",
        ) as handle:

            handle.write(
                content
            )

            handle.flush()

            os.fsync(
                handle.fileno()
            )

        os.replace(
            temp_name,
            path,
        )

    except Exception:

        try:
            os.unlink(
                temp_name
            )
        except FileNotFoundError:
            pass

        raise


# =========================================================
# CHECKPOINT BUILD / VALIDATE
# =========================================================

def cell33_build_checkpoint(
    request,
    result,
):

    checkpoint = {
        "checkpoint_version": (
            CELL33_RUNNER_VERSION
        ),

        "candidate_id": (
            request[
                "candidate_id"
            ]
        ),

        "request_fingerprint": (
            request[
                "request_fingerprint"
            ]
        ),

        "protocol_version": (
            GENERATION_V3_VERSION
        ),

        "protocol_sha256": (
            GENERATION_V3_PROTOCOL_SHA256
        ),

        "prompt_version": (
            CELL33_PROMPT_VERSION
        ),

        "output_schema_version": (
            CELL33_OUTPUT_SCHEMA_VERSION
        ),

        "model": (
            semantic_review_model
        ),

        "result": (
            result
        ),
    }

    checkpoint[
        "checkpoint_payload_sha256"
    ] = cell33_sha256_text(
        cell33_canonical_json(
            checkpoint
        )
    )

    return checkpoint


def cell33_validate_checkpoint(
    checkpoint,
    request,
):

    errors = []

    if not isinstance(checkpoint, dict):
        return [
            "CHECKPOINT_NOT_OBJECT"
        ]

    stored_sha = checkpoint.get(
        "checkpoint_payload_sha256"
    )

    payload_without_sha = {
        key: value
        for key, value in checkpoint.items()
        if key
        != "checkpoint_payload_sha256"
    }

    recalculated_sha = (
        cell33_sha256_text(
            cell33_canonical_json(
                payload_without_sha
            )
        )
    )

    if stored_sha != recalculated_sha:
        errors.append(
            "CHECKPOINT_HASH_MISMATCH"
        )

    exact_expectations = {
        "checkpoint_version": (
            CELL33_RUNNER_VERSION
        ),

        "candidate_id": (
            request[
                "candidate_id"
            ]
        ),

        "request_fingerprint": (
            request[
                "request_fingerprint"
            ]
        ),

        "protocol_version": (
            GENERATION_V3_VERSION
        ),

        "protocol_sha256": (
            GENERATION_V3_PROTOCOL_SHA256
        ),

        "prompt_version": (
            CELL33_PROMPT_VERSION
        ),

        "output_schema_version": (
            CELL33_OUTPUT_SCHEMA_VERSION
        ),

        "model": (
            semantic_review_model
        ),
    }

    for (
        field_name,
        expected_value,
    ) in exact_expectations.items():

        if (
            checkpoint.get(
                field_name
            )
            != expected_value
        ):
            errors.append(
                f"{field_name.upper()}_MISMATCH"
            )

    result = checkpoint.get(
        "result"
    )

    if not isinstance(result, dict):

        errors.append(
            "CHECKPOINT_RESULT_INVALID"
        )

    elif (
        result.get(
            "candidate_id"
        )
        != request[
            "candidate_id"
        ]
    ):

        errors.append(
            "CHECKPOINT_RESULT_CANDIDATE_MISMATCH"
        )

    return errors


def cell33_load_checkpoint(
    request,
):

    path = cell33_checkpoint_path(
        request[
            "candidate_id"
        ]
    )

    if not path.exists():

        return (
            None,
            "MISSING",
            [],
        )

    try:

        with path.open(
            "r",
            encoding="utf-8",
        ) as handle:

            checkpoint = json.load(
                handle
            )

    except Exception as exc:

        return (
            None,
            "UNREADABLE",
            [
                f"{type(exc).__name__}: {exc}"
            ],
        )

    errors = (
        cell33_validate_checkpoint(
            checkpoint,
            request,
        )
    )

    if errors:

        return (
            checkpoint,
            "INVALID",
            errors,
        )

    return (
        checkpoint,
        "VALID",
        [],
    )


# =========================================================
# COST
# =========================================================

cell33_has_cost_rates = (
    "INPUT_TOMAN_PER_TOKEN" in globals()
    and
    "OUTPUT_TOMAN_PER_TOKEN" in globals()
)


def cell33_estimated_cost_toman(
    input_tokens,
    output_tokens,
):

    if not cell33_has_cost_rates:
        return None

    return (
        input_tokens
        * INPUT_TOMAN_PER_TOKEN
        +
        output_tokens
        * OUTPUT_TOMAN_PER_TOKEN
    )


# =========================================================
# API EXECUTION
# =========================================================

def cell33_execute_request(
    request,
):

    attempt_history = []

    for attempt_number in range(
        1,
        CELL33_MAX_ATTEMPTS + 1,
    ):

        started_at = time.time()

        try:

            response = (
                semantic_review_client
                .chat
                .completions
                .create(
                    model=(
                        semantic_review_model
                    ),

                    messages=[
                        {
                            "role": "system",
                            "content": (
                                GENERATION_V3_SYSTEM_PROMPT
                            ),
                        },
                        {
                            "role": "user",
                            "content": (
                                request[
                                    "user_prompt"
                                ]
                            ),
                        },
                    ],

                    temperature=(
                        CELL33_TEMPERATURE
                    ),

                    top_p=(
                        CELL33_TOP_P
                    ),

                    max_tokens=(
                        CELL33_MAX_TOKENS
                    ),

                    extra_body={
                        "enable_thinking": False,
                    },
                )
            )

        except Exception as exc:

            attempt_history.append({
                "attempt": (
                    attempt_number
                ),

                "status": (
                    "API_ERROR"
                ),

                "error": (
                    f"{type(exc).__name__}: {exc}"
                ),
            })

            if (
                attempt_number
                < CELL33_MAX_ATTEMPTS
            ):

                wait_seconds = (
                    70.0
                    if cell33_is_rate_limit_exception(
                        exc
                    )
                    else (
                        CELL33_RETRY_SECONDS
                        * attempt_number
                    )
                )

                time.sleep(
                    wait_seconds
                )

                continue

            return {
                "candidate_id": (
                    request[
                        "candidate_id"
                    ]
                ),

                "execution_status": (
                    "API_FAILED"
                ),

                "attempt_history": (
                    attempt_history
                ),
            }

        elapsed_seconds = (
            time.time()
            - started_at
        )

        choice = response.choices[0]
        message = choice.message

        finish_reason = getattr(
            choice,
            "finish_reason",
            None,
        )

        raw_content = getattr(
            message,
            "content",
            None,
        )

        usage = getattr(
            response,
            "usage",
            None,
        )

        input_tokens = int(
            getattr(
                usage,
                "prompt_tokens",
                0,
            )
            or 0
        )

        output_tokens = int(
            getattr(
                usage,
                "completion_tokens",
                0,
            )
            or 0
        )

        total_tokens = int(
            getattr(
                usage,
                "total_tokens",
                0,
            )
            or (
                input_tokens
                + output_tokens
            )
        )

        reasoning_tokens = 0

        completion_details = getattr(
            usage,
            "completion_tokens_details",
            None,
        )

        if completion_details is not None:

            reasoning_tokens = int(
                getattr(
                    completion_details,
                    "reasoning_tokens",
                    0,
                )
                or 0
            )

        clean_content = (
            cell33_clean_model_json(
                raw_content
            )
        )

        parsed_json = None
        parse_error = None

        try:

            parsed_json = json.loads(
                clean_content
            )

        except Exception as exc:

            parse_error = (
                f"{type(exc).__name__}: {exc}"
            )

        if parsed_json is None:

            structural_errors = [
                "JSON_PARSE_FAILED"
            ]

        else:

            structural_errors = (
                cell33_validate_output(
                    parsed_json
                )
            )

        return {
            "candidate_id": (
                request[
                    "candidate_id"
                ]
            ),

            "source_seed_slot_id": (
                request[
                    "source_seed_slot_id"
                ]
            ),

            "claim_id": (
                request[
                    "claim_id"
                ]
            ),

            "claim_type": (
                request[
                    "claim_type"
                ]
            ),

            "evidence_count": (
                request[
                    "evidence_count"
                ]
            ),

            "request_fingerprint": (
                request[
                    "request_fingerprint"
                ]
            ),

            "model": (
                semantic_review_model
            ),

            "protocol_version": (
                GENERATION_V3_VERSION
            ),

            "protocol_sha256": (
                GENERATION_V3_PROTOCOL_SHA256
            ),

            "prompt_version": (
                CELL33_PROMPT_VERSION
            ),

            "output_schema_version": (
                CELL33_OUTPUT_SCHEMA_VERSION
            ),

            "attempt_number": (
                attempt_number
            ),

            "finish_reason": (
                finish_reason
            ),

            "elapsed_seconds": round(
                elapsed_seconds,
                4,
            ),

            "usage": {
                "input_tokens": (
                    input_tokens
                ),

                "output_tokens": (
                    output_tokens
                ),

                "total_tokens": (
                    total_tokens
                ),

                "reasoning_tokens": (
                    reasoning_tokens
                ),
            },

            "estimated_cost_toman": (
                cell33_estimated_cost_toman(
                    input_tokens,
                    output_tokens,
                )
            ),

            "raw_content": (
                raw_content
            ),

            "parsed_json": (
                parsed_json
            ),

            "parse_error": (
                parse_error
            ),

            "structural_errors": (
                structural_errors
            ),

            "structural_status": (
                "PASS"
                if not structural_errors
                else "FAIL"
            ),

            "attempt_history": (
                attempt_history
            ),
        }

    raise RuntimeError(
        "Unexpected Cell 33 execution path."
    )


# =========================================================
# PREFLIGHT
# =========================================================

print(
    "=== GOLDEN V2 QA GENERATION V3 CANARY PREFLIGHT ==="
)

print()

print(
    "Canary population:",
    len(
        cell33_canary_requests
    ),
)

print(
    "Model:",
    semantic_review_model,
)

print(
    "Protocol:",
    GENERATION_V3_VERSION,
)

print(
    "Protocol SHA-256:",
    GENERATION_V3_PROTOCOL_SHA256,
)

print(
    "Checkpoint directory:",
    cell33_checkpoint_dir.relative_to(
        Path(project_root)
    ),
)

print(
    "API allowed for missing exact checkpoints:",
    CELL33_ALLOW_API_FOR_MISSING,
)

print(
    "Full Generation enabled:",
    False,
)

print()


# =========================================================
# RESOLVE CANARY
# =========================================================

cell33_canary_results = []
cell33_resolution_modes = []

cell33_api_calls_this_execution = 0
cell33_checkpoints_created_this_execution = 0


for index, request in enumerate(
    cell33_canary_requests,
    start=1,
):

    (
        checkpoint,
        checkpoint_status,
        checkpoint_errors,
    ) = cell33_load_checkpoint(
        request
    )

    candidate_id = request[
        "candidate_id"
    ]

    if checkpoint_status == "VALID":

        result = checkpoint[
            "result"
        ]

        resolution_mode = (
            "EXACT_CHECKPOINT"
        )

    else:

        if checkpoint_status in {
            "INVALID",
            "UNREADABLE",
        }:

            raise AssertionError(
                "Existing Cell 33 checkpoint is invalid. "
                "Refusing automatic paid replacement for "
                f"{candidate_id}. "
                f"Status={checkpoint_status}; "
                f"errors={checkpoint_errors}"
            )

        if not CELL33_ALLOW_API_FOR_MISSING:

            raise AssertionError(
                "Exact Cell 33 checkpoint is missing and "
                "API execution is disabled for "
                f"{candidate_id}."
            )

        if (
            cell33_api_calls_this_execution
            > 0
        ):

            time.sleep(
                CELL33_REQUEST_SPACING_SECONDS
            )

        result = (
            cell33_execute_request(
                request
            )
        )

        cell33_api_calls_this_execution += 1

        checkpoint = (
            cell33_build_checkpoint(
                request,
                result,
            )
        )

        checkpoint_path = (
            cell33_checkpoint_path(
                candidate_id
            )
        )

        cell33_atomic_write_json(
            checkpoint_path,
            checkpoint,
        )

        (
            verified_checkpoint,
            verified_status,
            verified_errors,
        ) = cell33_load_checkpoint(
            request
        )

        assert (
            verified_status
            == "VALID"
        ), (
            "New Cell 33 checkpoint verification failed "
            f"for {candidate_id}: "
            f"{verified_errors}"
        )

        result = (
            verified_checkpoint[
                "result"
            ]
        )

        resolution_mode = (
            "NEW_PAID_EXECUTION"
        )

        cell33_checkpoints_created_this_execution += 1

    cell33_canary_results.append(
        result
    )

    cell33_resolution_modes.append({
        "candidate_id": (
            candidate_id
        ),

        "resolution_mode": (
            resolution_mode
        ),
    })

    print(
        f"[{index:02d}/12]",
        candidate_id,
        "|",
        resolution_mode,
        "|",
        result.get(
            "structural_status",
            result.get(
                "execution_status"
            ),
        ),
        flush=True,
    )


# =========================================================
# FINAL DISK RESCAN
# =========================================================

cell33_final_valid_checkpoint_count = 0
cell33_final_missing_checkpoint_count = 0
cell33_final_invalid_checkpoint_count = 0


for request in cell33_canary_requests:

    (
        checkpoint,
        status,
        errors,
    ) = cell33_load_checkpoint(
        request
    )

    if status == "VALID":

        cell33_final_valid_checkpoint_count += 1

    elif status == "MISSING":

        cell33_final_missing_checkpoint_count += 1

    else:

        cell33_final_invalid_checkpoint_count += 1


# =========================================================
# SUMMARY
# =========================================================

cell33_structural_pass_count = sum(
    result.get(
        "structural_status"
    )
    == "PASS"
    for result in cell33_canary_results
)

cell33_api_failure_count = sum(
    result.get(
        "execution_status"
    )
    == "API_FAILED"
    for result in cell33_canary_results
)

cell33_total_input_tokens = sum(
    (
        result.get("usage")
        or {}
    ).get(
        "input_tokens",
        0,
    )
    for result in cell33_canary_results
)

cell33_total_output_tokens = sum(
    (
        result.get("usage")
        or {}
    ).get(
        "output_tokens",
        0,
    )
    for result in cell33_canary_results
)

cell33_total_tokens = sum(
    (
        result.get("usage")
        or {}
    ).get(
        "total_tokens",
        0,
    )
    for result in cell33_canary_results
)

cell33_total_reasoning_tokens = sum(
    (
        result.get("usage")
        or {}
    ).get(
        "reasoning_tokens",
        0,
    )
    for result in cell33_canary_results
)

cell33_known_costs = [
    result.get(
        "estimated_cost_toman"
    )
    for result in cell33_canary_results
    if isinstance(
        result.get(
            "estimated_cost_toman"
        ),
        (int, float),
    )
]

cell33_total_estimated_cost_toman = (
    sum(cell33_known_costs)
    if cell33_known_costs
    else None
)


# =========================================================
# RUNNER + EXPERIMENTAL STATUS
# =========================================================

cell33_runner_checks = {
    "12_results_reconstructed": (
        len(cell33_canary_results)
        == 12
    ),

    "12_exact_valid_checkpoints": (
        cell33_final_valid_checkpoint_count
        == 12
    ),

    "zero_missing_checkpoints": (
        cell33_final_missing_checkpoint_count
        == 0
    ),

    "zero_invalid_checkpoints": (
        cell33_final_invalid_checkpoint_count
        == 0
    ),

    "zero_api_failures": (
        cell33_api_failure_count
        == 0
    ),

    "full_generation_not_started": True,
}

cell33_runner_status = (
    "PASS"
    if all(
        cell33_runner_checks.values()
    )
    else "FAIL"
)

cell33_experimental_status = (
    "PASS"
    if cell33_structural_pass_count
    == 12
    else "FAIL"
)

cell33_next_readiness = (
    "READY_FOR_QA_GENERATION_V3_SEMANTIC_CANARY_VALIDATION"
    if (
        cell33_runner_status == "PASS"
        and cell33_experimental_status == "PASS"
    )
    else (
        "READY_FOR_CELL33_FAILURE_ANALYSIS"
        if cell33_runner_status == "PASS"
        else "BLOCKED_CELL33_RUNNER_REPAIR_REQUIRED"
    )
)


# =========================================================
# OUTPUT
# =========================================================

print()

print(
    "=== GOLDEN V2 QA GENERATION V3 CANARY RESULT ==="
)

print()

print(
    "Results reconstructed:",
    len(
        cell33_canary_results
    ),
)

print(
    "Exact valid checkpoints:",
    cell33_final_valid_checkpoint_count,
)

print(
    "Missing checkpoints:",
    cell33_final_missing_checkpoint_count,
)

print(
    "Invalid checkpoints:",
    cell33_final_invalid_checkpoint_count,
)

print(
    "Resolution modes:",
    dict(
        Counter(
            row[
                "resolution_mode"
            ]
            for row in cell33_resolution_modes
        )
    ),
)

print(
    "Checkpoints created this execution:",
    cell33_checkpoints_created_this_execution,
)

print(
    "API calls this execution:",
    cell33_api_calls_this_execution,
)

print()

print(
    "Structural PASS:",
    cell33_structural_pass_count,
    "/ 12",
)

print(
    "API failures:",
    cell33_api_failure_count,
)

print()

print(
    "Input tokens:",
    cell33_total_input_tokens,
)

print(
    "Output tokens:",
    cell33_total_output_tokens,
)

print(
    "Total tokens:",
    cell33_total_tokens,
)

print(
    "Reasoning tokens:",
    cell33_total_reasoning_tokens,
)

print(
    "Estimated total cost (toman):",
    cell33_total_estimated_cost_toman,
)

print()

print(
    "Runner checks:"
)

for (
    check_name,
    check_value,
) in cell33_runner_checks.items():

    print(
        " -",
        check_name,
        ":",
        (
            "PASS"
            if check_value
            else "FAIL"
        ),
    )

print()

print(
    "Runner health:",
    cell33_runner_status,
)

print(
    "Experimental structural result:",
    cell33_experimental_status,
)

print(
    "Full Generation started:",
    False,
)

print(
    "Next readiness:",
    cell33_next_readiness,
)

print(
    "CELL 33 CHECKPOINTED RUNNER:",
    cell33_runner_status,
)

assert cell33_runner_status == "PASS", (
    "Cell 33 checkpointed runner health failed."
)
````

<h2>مرحله ۳۴ — تحلیل شکست‌های Canary در Generation v3</h2>
<p>
در این مرحله چهار خروجی Canary که از قرارداد ساختاری Generation v3 عبور نکرده‌اند بررسی می‌شوند.
هدف تشخیص دقیق نوع شکست است: خطای JSON، فیلدهای نامعتبر، پاسخ ناقص، ارجاع نامناسب به Context یا سایر تخطی‌های ساختاری.
این مرحله فقط نتایج موجود Cell 33 را تحلیل می‌کند، هیچ فراخوانی API انجام نمی‌دهد و هیچ داده‌ای را تغییر نمی‌دهد.
</p>

````python
# Code Cell 34 — Generation v3 Canary Failure Analysis

import json
from collections import Counter


# =========================================================
# REQUIRED STATE
# =========================================================

required_cell34_names = [
    "cell33_canary_results",
    "cell33_skeleton_by_candidate_id",
]

missing_cell34_names = [
    name
    for name in required_cell34_names
    if name not in globals()
]

assert not missing_cell34_names, (
    "Missing required Cell 33 state: "
    + ", ".join(missing_cell34_names)
)

assert len(cell33_canary_results) == 12


# =========================================================
# SPLIT PASS / FAIL
# =========================================================

cell34_structural_pass = [
    result
    for result in cell33_canary_results
    if result.get("structural_status") == "PASS"
]

cell34_structural_fail = [
    result
    for result in cell33_canary_results
    if result.get("structural_status") != "PASS"
]

assert len(cell34_structural_pass) == 8
assert len(cell34_structural_fail) == 4


# =========================================================
# FAILURE DISTRIBUTION
# =========================================================

cell34_failure_reason_distribution = Counter()

for result in cell34_structural_fail:

    errors = result.get(
        "structural_errors"
    ) or []

    if not errors:
        cell34_failure_reason_distribution[
            "NO_STRUCTURAL_ERROR_RECORDED"
        ] += 1

    else:
        for error in errors:
            cell34_failure_reason_distribution[
                error
            ] += 1


# =========================================================
# DETAILED FAILURE AUDIT
# =========================================================

cell34_failure_audit = []

for result in cell34_structural_fail:

    candidate_id = result[
        "candidate_id"
    ]

    skeleton = (
        cell33_skeleton_by_candidate_id[
            candidate_id
        ]
    )

    parsed_json = result.get(
        "parsed_json"
    )

    raw_content = result.get(
        "raw_content"
    )

    parse_error = result.get(
        "parse_error"
    )

    structural_errors = (
        result.get(
            "structural_errors"
        )
        or []
    )

    if isinstance(parsed_json, dict):
        observed_fields = sorted(
            parsed_json.keys()
        )
    else:
        observed_fields = []

    audit_record = {
        "candidate_id": (
            candidate_id
        ),

        "claim_id": (
            skeleton[
                "claim_id"
            ]
        ),

        "claim_text_fa": (
            skeleton[
                "claim_text_fa"
            ]
        ),

        "evidence_quotes": (
            skeleton[
                "evidence_quotes"
            ]
        ),

        "finish_reason": (
            result.get(
                "finish_reason"
            )
        ),

        "parse_error": (
            parse_error
        ),

        "structural_errors": (
            structural_errors
        ),

        "observed_fields": (
            observed_fields
        ),

        "parsed_json": (
            parsed_json
        ),

        "raw_content": (
            raw_content
        ),
    }

    cell34_failure_audit.append(
        audit_record
    )


# =========================================================
# CLASSIFY FAILURE SHAPE
# =========================================================

cell34_failure_classes = Counter()

for audit in cell34_failure_audit:

    errors = set(
        audit[
            "structural_errors"
        ]
    )

    if "JSON_PARSE_FAILED" in errors:

        failure_class = (
            "JSON_OR_TRUNCATION_FAILURE"
        )

    elif "OUTPUT_FIELDS_MISMATCH" in errors:

        failure_class = (
            "OUTPUT_SCHEMA_FAILURE"
        )

    elif "QUESTION_CONTAINS_CONTEXT_REFERENCE" in errors:

        failure_class = (
            "QUESTION_CONTEXT_REFERENCE_FAILURE"
        )

    elif errors:

        failure_class = (
            "FIELD_CONTENT_FAILURE"
        )

    else:

        failure_class = (
            "UNKNOWN_FAILURE"
        )

    audit[
        "failure_class"
    ] = failure_class

    cell34_failure_classes[
        failure_class
    ] += 1


# =========================================================
# SAFETY / STATUS
# =========================================================

cell34_checks = {
    "12_canary_results_available": (
        len(cell33_canary_results) == 12
    ),

    "8_structural_pass_confirmed": (
        len(cell34_structural_pass) == 8
    ),

    "4_structural_fail_confirmed": (
        len(cell34_structural_fail) == 4
    ),

    "4_failures_audited": (
        len(cell34_failure_audit) == 4
    ),

    "every_failure_classified": (
        sum(
            cell34_failure_classes.values()
        )
        == 4
    ),

    "zero_api_calls": True,

    "zero_file_writes": True,
}

cell34_status = (
    "PASS"
    if all(
        cell34_checks.values()
    )
    else "FAIL"
)

cell34_next_readiness = (
    "READY_FOR_GENERATION_V3_FAILURE_ADJUDICATION"
    if cell34_status == "PASS"
    else "BLOCKED_CELL34_REPAIR_REQUIRED"
)


# =========================================================
# OUTPUT
# =========================================================

print(
    "=== GOLDEN V2 GENERATION V3 CANARY FAILURE ANALYSIS ==="
)

print()

print(
    "Canary population:",
    len(cell33_canary_results),
)

print(
    "Structural PASS:",
    len(cell34_structural_pass),
)

print(
    "Structural FAIL:",
    len(cell34_structural_fail),
)

print()

print(
    "Failure reason distribution:"
)

for reason, count in sorted(
    cell34_failure_reason_distribution.items()
):
    print(
        " -",
        reason,
        ":",
        count,
    )

print()

print(
    "Failure class distribution:"
)

for failure_class, count in sorted(
    cell34_failure_classes.items()
):
    print(
        " -",
        failure_class,
        ":",
        count,
    )


# =========================================================
# PRINT ALL FOUR FAILURES
# =========================================================

for index, audit in enumerate(
    cell34_failure_audit,
    start=1,
):

    print()
    print(
        "=" * 70
    )

    print(
        f"FAILURE {index} / 4"
    )

    print(
        "=" * 70
    )

    print(
        "Candidate ID:",
        audit[
            "candidate_id"
        ],
    )

    print(
        "Claim ID:",
        audit[
            "claim_id"
        ],
    )

    print(
        "Failure class:",
        audit[
            "failure_class"
        ],
    )

    print(
        "Structural errors:",
        audit[
            "structural_errors"
        ],
    )

    print(
        "Finish reason:",
        audit[
            "finish_reason"
        ],
    )

    print(
        "Parse error:",
        audit[
            "parse_error"
        ],
    )

    print(
        "Observed fields:",
        audit[
            "observed_fields"
        ],
    )

    print()

    print(
        "CLAIM:"
    )

    print(
        audit[
            "claim_text_fa"
        ]
    )

    print()

    print(
        "EVIDENCE:"
    )

    for evidence_index, quote in enumerate(
        audit[
            "evidence_quotes"
        ],
        start=1,
    ):

        print(
            f"[{evidence_index}]",
            quote,
        )

    print()

    print(
        "PARSED JSON:"
    )

    print(
        json.dumps(
            audit[
                "parsed_json"
            ],
            ensure_ascii=False,
            indent=2,
        )
    )

    print()

    print(
        "RAW MODEL OUTPUT:"
    )

    print(
        audit[
            "raw_content"
        ]
    )


# =========================================================
# FINAL CHECKS
# =========================================================

print()
print(
    "=" * 70
)

print(
    "Validation checks:"
)

for (
    check_name,
    check_value,
) in cell34_checks.items():

    print(
        " -",
        check_name,
        ":",
        (
            "PASS"
            if check_value
            else "FAIL"
        ),
    )

print()

print(
    "API calls performed:",
    0,
)

print(
    "Files modified:",
    0,
)

print(
    "CELL 34 STATUS:",
    cell34_status,
)

print(
    "Next readiness:",
    cell34_next_readiness,
)

assert cell34_status == "PASS", (
    "Cell 34 failure analysis failed."
)
````

تولید کامل ۳۸۱۳ رکورد همچنان آغاز نمی‌شود.

````python
# Code Cell 35 — QA Generation v3 Canary 768-Token Checkpointed Execution

from collections import Counter
from pathlib import Path
import hashlib
import json
import os
import re
import tempfile
import time


# =========================================================
# CONFIG
# =========================================================

CELL35_RUNNER_VERSION = (
    "golden_v2_generation_canary_v3_checkpoint_runner_v2"
)

CELL35_PROMPT_VERSION = (
    "golden_v2_qa_generation_prompt_v3"
)

CELL35_OUTPUT_SCHEMA_VERSION = (
    "golden_v2_qa_generation_output_v3"
)

CELL35_MAX_TOKENS = 768
CELL35_TEMPERATURE = 0.7
CELL35_TOP_P = 0.8

CELL35_MAX_ATTEMPTS = 2
CELL35_RETRY_SECONDS = 2.0
CELL35_REQUEST_SPACING_SECONDS = 2.7

# Intentional paid Canary execution.
# MUST become False before final Restart -> Run All.
CELL35_ALLOW_API_FOR_MISSING = False


# =========================================================
# REQUIRED STATE
# =========================================================

required_cell35_names = [
    "project_root",
    "semantic_review_client",
    "semantic_review_model",
    "cell25_candidate_skeletons",
    "cell26_canary_requests",
    "GENERATION_V3_VERSION",
    "GENERATION_V3_SYSTEM_PROMPT",
    "GENERATION_V3_PROTOCOL_SHA256",
]

missing_cell35_names = [
    name
    for name in required_cell35_names
    if name not in globals()
]

assert not missing_cell35_names, (
    "Missing required Cell 35 state: "
    + ", ".join(missing_cell35_names)
)

assert len(cell25_candidate_skeletons) == 3813
assert len(cell26_canary_requests) == 12

assert (
    semantic_review_model.lower()
    == "qwen3-30b-a3b".lower()
), (
    "Unexpected semantic_review_model: "
    f"{semantic_review_model}"
)


# =========================================================
# LOOKUPS
# =========================================================

cell35_skeleton_by_candidate_id = {
    record["candidate_id"]: record
    for record in cell25_candidate_skeletons
}

assert len(cell35_skeleton_by_candidate_id) == 3813

cell35_canary_candidate_ids = [
    request["candidate_id"]
    for request in cell26_canary_requests
]

assert len(cell35_canary_candidate_ids) == 12
assert len(set(cell35_canary_candidate_ids)) == 12

missing_canary_candidates = [
    candidate_id
    for candidate_id in cell35_canary_candidate_ids
    if candidate_id not in cell35_skeleton_by_candidate_id
]

assert not missing_canary_candidates, (
    "Canary candidates missing from Cell 25 skeletons: "
    + ", ".join(missing_canary_candidates)
)


# =========================================================
# CANONICAL JSON / HASH
# =========================================================

def cell35_canonical_json(value):

    return json.dumps(
        value,
        ensure_ascii=False,
        sort_keys=True,
        separators=(",", ":"),
    )


def cell35_sha256_text(text):

    return hashlib.sha256(
        text.encode("utf-8")
    ).hexdigest()


# =========================================================
# USER PAYLOAD / PROMPT
# =========================================================

def cell35_build_payload(
    skeleton,
):

    evidence = []

    for (
        evidence_id,
        evidence_quote,
        evidence_locator,
    ) in zip(
        skeleton["evidence_ids"],
        skeleton["evidence_quotes"],
        skeleton["evidence_locators"],
    ):

        evidence.append({
            "evidence_id": evidence_id,
            "quote": evidence_quote,
            "locator": evidence_locator,
        })

    return {
        "candidate_id": skeleton["candidate_id"],

        "target_claim": {
            "claim_id": skeleton["claim_id"],
            "text_fa": skeleton["claim_text_fa"],
            "claim_type": skeleton["claim_type"],
            "claim_types": skeleton["claim_types"],
            "authority_class": skeleton["authority_class"],
        },

        "evidence": evidence,

        "generation_instruction": (
            "Generate exactly one Persian standalone question "
            "and one Persian expected answer that directly "
            "evaluate the target Claim. Evidence is grounding "
            "context only. Do not re-adjudicate Claim support."
        ),

        "required_output_fields": [
            "question",
            "expected_answer",
            "question_intent",
        ],
    }


def cell35_build_user_prompt(
    skeleton,
):

    payload = cell35_build_payload(
        skeleton
    )

    return (
        "Generate the requested Golden Dataset evaluation "
        "question and expected answer from this JSON input.\n\n"
        + json.dumps(
            payload,
            ensure_ascii=False,
            sort_keys=True,
            indent=2,
        )
    )


# =========================================================
# OUTPUT VALIDATION
# =========================================================

def cell35_validate_output(
    parsed_json,
):

    errors = []

    if not isinstance(parsed_json, dict):
        return ["OUTPUT_NOT_OBJECT"]

    expected_fields = {
        "question",
        "expected_answer",
        "question_intent",
    }

    actual_fields = set(
        parsed_json.keys()
    )

    if actual_fields != expected_fields:
        errors.append(
            "OUTPUT_FIELDS_MISMATCH"
        )

    for field_name in expected_fields:

        value = parsed_json.get(
            field_name
        )

        if (
            not isinstance(value, str)
            or not value.strip()
        ):
            errors.append(
                f"INVALID_{field_name.upper()}"
            )

    question = parsed_json.get(
        "question"
    )

    if isinstance(question, str):

        forbidden_question_phrases = [
            "طبق متن بالا",
            "در متن بالا",
            "طبق شواهد",
            "بر اساس شواهد",
            "طبق claim",
            "طبق evidence",
            "در dataset",
        ]

        question_lower = question.lower()

        for phrase in forbidden_question_phrases:

            if phrase.lower() in question_lower:

                errors.append(
                    "QUESTION_CONTAINS_CONTEXT_REFERENCE"
                )

                break

    return sorted(
        set(errors)
    )


# =========================================================
# RESPONSE CLEANING
# =========================================================

def cell35_clean_model_json(
    content,
):

    if not isinstance(content, str):
        return ""

    clean = content.strip()

    if clean.startswith("```"):

        clean = re.sub(
            r"^```(?:json)?\s*",
            "",
            clean,
            flags=re.IGNORECASE,
        )

        clean = re.sub(
            r"\s*```$",
            "",
            clean,
        )

    return clean.strip()


def cell35_is_rate_limit_exception(
    exc,
):

    error_text = str(exc).lower()
    error_type = type(exc).__name__.lower()

    return (
        "ratelimit" in error_type
        or "rate_limit" in error_type
        or "429" in error_text
        or "too many requests" in error_text
    )


# =========================================================
# REQUEST CONTRACT
# =========================================================

def cell35_build_request(
    skeleton,
):

    candidate_id = skeleton[
        "candidate_id"
    ]

    user_prompt = (
        cell35_build_user_prompt(
            skeleton
        )
    )

    fingerprint_payload = {
        "runner_version": (
            CELL35_RUNNER_VERSION
        ),

        "candidate_id": (
            candidate_id
        ),

        "model": (
            semantic_review_model
        ),

        "protocol_version": (
            GENERATION_V3_VERSION
        ),

        "protocol_sha256": (
            GENERATION_V3_PROTOCOL_SHA256
        ),

        "prompt_version": (
            CELL35_PROMPT_VERSION
        ),

        "output_schema_version": (
            CELL35_OUTPUT_SCHEMA_VERSION
        ),

        "system_prompt": (
            GENERATION_V3_SYSTEM_PROMPT
        ),

        "user_prompt": (
            user_prompt
        ),

        "temperature": (
            CELL35_TEMPERATURE
        ),

        "top_p": (
            CELL35_TOP_P
        ),

        "max_tokens": (
            CELL35_MAX_TOKENS
        ),

        "enable_thinking": False,
    }

    request_fingerprint = (
        cell35_sha256_text(
            cell35_canonical_json(
                fingerprint_payload
            )
        )
    )

    return {
        "candidate_id": (
            candidate_id
        ),

        "source_seed_slot_id": (
            skeleton[
                "source_seed_slot_id"
            ]
        ),

        "claim_id": (
            skeleton[
                "claim_id"
            ]
        ),

        "claim_type": (
            skeleton[
                "claim_type"
            ]
        ),

        "evidence_count": len(
            skeleton[
                "evidence_ids"
            ]
        ),

        "user_prompt": (
            user_prompt
        ),

        "request_fingerprint": (
            request_fingerprint
        ),
    }


cell35_canary_requests = [
    cell35_build_request(
        cell35_skeleton_by_candidate_id[
            candidate_id
        ]
    )
    for candidate_id in (
        cell35_canary_candidate_ids
    )
]

assert len(cell35_canary_requests) == 12


# =========================================================
# CHECKPOINT DIRECTORY
# =========================================================

cell35_checkpoint_dir = (
    Path(project_root)
    / "artifacts"
    / "phase2"
    / "golden_dataset_v2"
    / "work"
    / "generation_canary_v3_retry_768"
)

cell35_checkpoint_dir.mkdir(
    parents=True,
    exist_ok=True,
)


def cell35_checkpoint_path(
    candidate_id,
):

    safe_id = re.sub(
        r"[^A-Za-z0-9_.-]+",
        "_",
        candidate_id,
    )

    return (
        cell35_checkpoint_dir
        / f"{safe_id}.json"
    )


# =========================================================
# ATOMIC CHECKPOINT WRITE
# =========================================================

def cell35_atomic_write_json(
    path,
    payload,
):

    path = Path(path)

    path.parent.mkdir(
        parents=True,
        exist_ok=True,
    )

    content = (
        json.dumps(
            payload,
            ensure_ascii=False,
            sort_keys=True,
            indent=2,
        )
        + "\n"
    )

    fd, temp_name = tempfile.mkstemp(
        prefix=path.name + ".",
        suffix=".tmp",
        dir=str(path.parent),
    )

    try:

        with os.fdopen(
            fd,
            "w",
            encoding="utf-8",
        ) as handle:

            handle.write(
                content
            )

            handle.flush()

            os.fsync(
                handle.fileno()
            )

        os.replace(
            temp_name,
            path,
        )

    except Exception:

        try:
            os.unlink(
                temp_name
            )
        except FileNotFoundError:
            pass

        raise


# =========================================================
# CHECKPOINT BUILD / VALIDATE
# =========================================================

def cell35_build_checkpoint(
    request,
    result,
):

    checkpoint = {
        "checkpoint_version": (
            CELL35_RUNNER_VERSION
        ),

        "candidate_id": (
            request[
                "candidate_id"
            ]
        ),

        "request_fingerprint": (
            request[
                "request_fingerprint"
            ]
        ),

        "protocol_version": (
            GENERATION_V3_VERSION
        ),

        "protocol_sha256": (
            GENERATION_V3_PROTOCOL_SHA256
        ),

        "prompt_version": (
            CELL35_PROMPT_VERSION
        ),

        "output_schema_version": (
            CELL35_OUTPUT_SCHEMA_VERSION
        ),

        "model": (
            semantic_review_model
        ),

        "result": (
            result
        ),
    }

    checkpoint[
        "checkpoint_payload_sha256"
    ] = cell35_sha256_text(
        cell35_canonical_json(
            checkpoint
        )
    )

    return checkpoint


def cell35_validate_checkpoint(
    checkpoint,
    request,
):

    errors = []

    if not isinstance(checkpoint, dict):
        return [
            "CHECKPOINT_NOT_OBJECT"
        ]

    stored_sha = checkpoint.get(
        "checkpoint_payload_sha256"
    )

    payload_without_sha = {
        key: value
        for key, value in checkpoint.items()
        if key != "checkpoint_payload_sha256"
    }

    recalculated_sha = (
        cell35_sha256_text(
            cell35_canonical_json(
                payload_without_sha
            )
        )
    )

    if stored_sha != recalculated_sha:
        errors.append(
            "CHECKPOINT_HASH_MISMATCH"
        )

    exact_expectations = {
        "checkpoint_version": (
            CELL35_RUNNER_VERSION
        ),

        "candidate_id": (
            request[
                "candidate_id"
            ]
        ),

        "request_fingerprint": (
            request[
                "request_fingerprint"
            ]
        ),

        "protocol_version": (
            GENERATION_V3_VERSION
        ),

        "protocol_sha256": (
            GENERATION_V3_PROTOCOL_SHA256
        ),

        "prompt_version": (
            CELL35_PROMPT_VERSION
        ),

        "output_schema_version": (
            CELL35_OUTPUT_SCHEMA_VERSION
        ),

        "model": (
            semantic_review_model
        ),
    }

    for (
        field_name,
        expected_value,
    ) in exact_expectations.items():

        if (
            checkpoint.get(
                field_name
            )
            != expected_value
        ):
            errors.append(
                f"{field_name.upper()}_MISMATCH"
            )

    result = checkpoint.get(
        "result"
    )

    if not isinstance(result, dict):

        errors.append(
            "CHECKPOINT_RESULT_INVALID"
        )

    elif (
        result.get(
            "candidate_id"
        )
        != request[
            "candidate_id"
        ]
    ):

        errors.append(
            "CHECKPOINT_RESULT_CANDIDATE_MISMATCH"
        )

    return errors


def cell35_load_checkpoint(
    request,
):

    path = cell35_checkpoint_path(
        request[
            "candidate_id"
        ]
    )

    if not path.exists():

        return (
            None,
            "MISSING",
            [],
        )

    try:

        with path.open(
            "r",
            encoding="utf-8",
        ) as handle:

            checkpoint = json.load(
                handle
            )

    except Exception as exc:

        return (
            None,
            "UNREADABLE",
            [
                f"{type(exc).__name__}: {exc}"
            ],
        )

    errors = (
        cell35_validate_checkpoint(
            checkpoint,
            request,
        )
    )

    if errors:

        return (
            checkpoint,
            "INVALID",
            errors,
        )

    return (
        checkpoint,
        "VALID",
        [],
    )


# =========================================================
# COST
# =========================================================

cell35_has_cost_rates = (
    "INPUT_TOMAN_PER_TOKEN" in globals()
    and
    "OUTPUT_TOMAN_PER_TOKEN" in globals()
)


def cell35_estimated_cost_toman(
    input_tokens,
    output_tokens,
):

    if not cell35_has_cost_rates:
        return None

    return (
        input_tokens
        * INPUT_TOMAN_PER_TOKEN
        +
        output_tokens
        * OUTPUT_TOMAN_PER_TOKEN
    )


# =========================================================
# API EXECUTION
# =========================================================

def cell35_execute_request(
    request,
):

    attempt_history = []

    for attempt_number in range(
        1,
        CELL35_MAX_ATTEMPTS + 1,
    ):

        started_at = time.time()

        try:

            response = (
                semantic_review_client
                .chat
                .completions
                .create(
                    model=(
                        semantic_review_model
                    ),

                    messages=[
                        {
                            "role": "system",
                            "content": (
                                GENERATION_V3_SYSTEM_PROMPT
                            ),
                        },
                        {
                            "role": "user",
                            "content": (
                                request[
                                    "user_prompt"
                                ]
                            ),
                        },
                    ],

                    temperature=(
                        CELL35_TEMPERATURE
                    ),

                    top_p=(
                        CELL35_TOP_P
                    ),

                    max_tokens=(
                        CELL35_MAX_TOKENS
                    ),

                    extra_body={
                        "enable_thinking": False,
                    },
                )
            )

        except Exception as exc:

            attempt_history.append({
                "attempt": (
                    attempt_number
                ),

                "status": (
                    "API_ERROR"
                ),

                "error": (
                    f"{type(exc).__name__}: {exc}"
                ),
            })

            if (
                attempt_number
                < CELL35_MAX_ATTEMPTS
            ):

                wait_seconds = (
                    70.0
                    if cell35_is_rate_limit_exception(
                        exc
                    )
                    else (
                        CELL35_RETRY_SECONDS
                        * attempt_number
                    )
                )

                time.sleep(
                    wait_seconds
                )

                continue

            return {
                "candidate_id": (
                    request[
                        "candidate_id"
                    ]
                ),

                "execution_status": (
                    "API_FAILED"
                ),

                "attempt_history": (
                    attempt_history
                ),
            }

        elapsed_seconds = (
            time.time()
            - started_at
        )

        choice = response.choices[0]
        message = choice.message

        finish_reason = getattr(
            choice,
            "finish_reason",
            None,
        )

        raw_content = getattr(
            message,
            "content",
            None,
        )

        usage = getattr(
            response,
            "usage",
            None,
        )

        input_tokens = int(
            getattr(
                usage,
                "prompt_tokens",
                0,
            )
            or 0
        )

        output_tokens = int(
            getattr(
                usage,
                "completion_tokens",
                0,
            )
            or 0
        )

        total_tokens = int(
            getattr(
                usage,
                "total_tokens",
                0,
            )
            or (
                input_tokens
                + output_tokens
            )
        )

        reasoning_tokens = 0

        completion_details = getattr(
            usage,
            "completion_tokens_details",
            None,
        )

        if completion_details is not None:

            reasoning_tokens = int(
                getattr(
                    completion_details,
                    "reasoning_tokens",
                    0,
                )
                or 0
            )

        clean_content = (
            cell35_clean_model_json(
                raw_content
            )
        )

        parsed_json = None
        parse_error = None

        try:

            parsed_json = json.loads(
                clean_content
            )

        except Exception as exc:

            parse_error = (
                f"{type(exc).__name__}: {exc}"
            )

        if parsed_json is None:

            structural_errors = [
                "JSON_PARSE_FAILED"
            ]

        else:

            structural_errors = (
                cell35_validate_output(
                    parsed_json
                )
            )

        return {
            "candidate_id": (
                request[
                    "candidate_id"
                ]
            ),

            "source_seed_slot_id": (
                request[
                    "source_seed_slot_id"
                ]
            ),

            "claim_id": (
                request[
                    "claim_id"
                ]
            ),

            "claim_type": (
                request[
                    "claim_type"
                ]
            ),

            "evidence_count": (
                request[
                    "evidence_count"
                ]
            ),

            "request_fingerprint": (
                request[
                    "request_fingerprint"
                ]
            ),

            "model": (
                semantic_review_model
            ),

            "protocol_version": (
                GENERATION_V3_VERSION
            ),

            "protocol_sha256": (
                GENERATION_V3_PROTOCOL_SHA256
            ),

            "prompt_version": (
                CELL35_PROMPT_VERSION
            ),

            "output_schema_version": (
                CELL35_OUTPUT_SCHEMA_VERSION
            ),

            "attempt_number": (
                attempt_number
            ),

            "finish_reason": (
                finish_reason
            ),

            "elapsed_seconds": round(
                elapsed_seconds,
                4,
            ),

            "usage": {
                "input_tokens": (
                    input_tokens
                ),

                "output_tokens": (
                    output_tokens
                ),

                "total_tokens": (
                    total_tokens
                ),

                "reasoning_tokens": (
                    reasoning_tokens
                ),
            },

            "estimated_cost_toman": (
                cell35_estimated_cost_toman(
                    input_tokens,
                    output_tokens,
                )
            ),

            "raw_content": (
                raw_content
            ),

            "parsed_json": (
                parsed_json
            ),

            "parse_error": (
                parse_error
            ),

            "structural_errors": (
                structural_errors
            ),

            "structural_status": (
                "PASS"
                if not structural_errors
                else "FAIL"
            ),

            "attempt_history": (
                attempt_history
            ),
        }

    raise RuntimeError(
        "Unexpected Cell 35 execution path."
    )


# =========================================================
# PREFLIGHT
# =========================================================

print(
    "=== GOLDEN V2 QA GENERATION V3 768-TOKEN CANARY PREFLIGHT ==="
)

print()

print(
    "Canary population:",
    len(
        cell35_canary_requests
    ),
)

print(
    "Model:",
    semantic_review_model,
)

print(
    "Protocol:",
    GENERATION_V3_VERSION,
)

print(
    "Protocol SHA-256:",
    GENERATION_V3_PROTOCOL_SHA256,
)

print(
    "Max tokens:",
    CELL35_MAX_TOKENS,
)

print(
    "Checkpoint directory:",
    cell35_checkpoint_dir.relative_to(
        Path(project_root)
    ),
)

print(
    "API allowed for missing exact checkpoints:",
    CELL35_ALLOW_API_FOR_MISSING,
)

print(
    "Full Generation enabled:",
    False,
)

print()


# =========================================================
# RESOLVE CANARY
# =========================================================

cell35_canary_results = []
cell35_resolution_modes = []

cell35_api_calls_this_execution = 0
cell35_checkpoints_created_this_execution = 0


for index, request in enumerate(
    cell35_canary_requests,
    start=1,
):

    (
        checkpoint,
        checkpoint_status,
        checkpoint_errors,
    ) = cell35_load_checkpoint(
        request
    )

    candidate_id = request[
        "candidate_id"
    ]

    if checkpoint_status == "VALID":

        result = checkpoint[
            "result"
        ]

        resolution_mode = (
            "EXACT_CHECKPOINT"
        )

    else:

        if checkpoint_status in {
            "INVALID",
            "UNREADABLE",
        }:

            raise AssertionError(
                "Existing Cell 35 checkpoint is invalid. "
                "Refusing automatic paid replacement for "
                f"{candidate_id}. "
                f"Status={checkpoint_status}; "
                f"errors={checkpoint_errors}"
            )

        if not CELL35_ALLOW_API_FOR_MISSING:

            raise AssertionError(
                "Exact Cell 35 checkpoint is missing and "
                "API execution is disabled for "
                f"{candidate_id}."
            )

        if (
            cell35_api_calls_this_execution
            > 0
        ):

            time.sleep(
                CELL35_REQUEST_SPACING_SECONDS
            )

        result = (
            cell35_execute_request(
                request
            )
        )

        cell35_api_calls_this_execution += 1

        checkpoint = (
            cell35_build_checkpoint(
                request,
                result,
            )
        )

        checkpoint_path = (
            cell35_checkpoint_path(
                candidate_id
            )
        )

        cell35_atomic_write_json(
            checkpoint_path,
            checkpoint,
        )

        (
            verified_checkpoint,
            verified_status,
            verified_errors,
        ) = cell35_load_checkpoint(
            request
        )

        assert (
            verified_status
            == "VALID"
        ), (
            "New Cell 35 checkpoint verification failed "
            f"for {candidate_id}: "
            f"{verified_errors}"
        )

        result = (
            verified_checkpoint[
                "result"
            ]
        )

        resolution_mode = (
            "NEW_PAID_EXECUTION"
        )

        cell35_checkpoints_created_this_execution += 1

    cell35_canary_results.append(
        result
    )

    cell35_resolution_modes.append({
        "candidate_id": (
            candidate_id
        ),

        "resolution_mode": (
            resolution_mode
        ),
    })

    print(
        f"[{index:02d}/12]",
        candidate_id,
        "|",
        resolution_mode,
        "|",
        result.get(
            "structural_status",
            result.get(
                "execution_status"
            ),
        ),
        flush=True,
    )


# =========================================================
# FINAL DISK RESCAN
# =========================================================

cell35_final_valid_checkpoint_count = 0
cell35_final_missing_checkpoint_count = 0
cell35_final_invalid_checkpoint_count = 0


for request in cell35_canary_requests:

    (
        checkpoint,
        status,
        errors,
    ) = cell35_load_checkpoint(
        request
    )

    if status == "VALID":

        cell35_final_valid_checkpoint_count += 1

    elif status == "MISSING":

        cell35_final_missing_checkpoint_count += 1

    else:

        cell35_final_invalid_checkpoint_count += 1


# =========================================================
# SUMMARY
# =========================================================

cell35_structural_pass_count = sum(
    result.get(
        "structural_status"
    )
    == "PASS"
    for result in cell35_canary_results
)

cell35_api_failure_count = sum(
    result.get(
        "execution_status"
    )
    == "API_FAILED"
    for result in cell35_canary_results
)

cell35_finish_reason_distribution = Counter(
    result.get(
        "finish_reason"
    )
    for result in cell35_canary_results
)

cell35_total_input_tokens = sum(
    (
        result.get("usage")
        or {}
    ).get(
        "input_tokens",
        0,
    )
    for result in cell35_canary_results
)

cell35_total_output_tokens = sum(
    (
        result.get("usage")
        or {}
    ).get(
        "output_tokens",
        0,
    )
    for result in cell35_canary_results
)

cell35_total_tokens = sum(
    (
        result.get("usage")
        or {}
    ).get(
        "total_tokens",
        0,
    )
    for result in cell35_canary_results
)

cell35_total_reasoning_tokens = sum(
    (
        result.get("usage")
        or {}
    ).get(
        "reasoning_tokens",
        0,
    )
    for result in cell35_canary_results
)

cell35_known_costs = [
    result.get(
        "estimated_cost_toman"
    )
    for result in cell35_canary_results
    if isinstance(
        result.get(
            "estimated_cost_toman"
        ),
        (int, float),
    )
]

cell35_total_estimated_cost_toman = (
    sum(cell35_known_costs)
    if cell35_known_costs
    else None
)


# =========================================================
# RUNNER + EXPERIMENTAL STATUS
# =========================================================

cell35_runner_checks = {
    "12_results_reconstructed": (
        len(cell35_canary_results)
        == 12
    ),

    "12_exact_valid_checkpoints": (
        cell35_final_valid_checkpoint_count
        == 12
    ),

    "zero_missing_checkpoints": (
        cell35_final_missing_checkpoint_count
        == 0
    ),

    "zero_invalid_checkpoints": (
        cell35_final_invalid_checkpoint_count
        == 0
    ),

    "zero_api_failures": (
        cell35_api_failure_count
        == 0
    ),

    "full_generation_not_started": True,
}

cell35_runner_status = (
    "PASS"
    if all(
        cell35_runner_checks.values()
    )
    else "FAIL"
)

cell35_experimental_status = (
    "PASS"
    if cell35_structural_pass_count
    == 12
    else "FAIL"
)

cell35_next_readiness = (
    "READY_FOR_QA_GENERATION_V3_SEMANTIC_CANARY_VALIDATION"
    if (
        cell35_runner_status == "PASS"
        and cell35_experimental_status == "PASS"
    )
    else (
        "READY_FOR_CELL35_FAILURE_ANALYSIS"
        if cell35_runner_status == "PASS"
        else "BLOCKED_CELL35_RUNNER_REPAIR_REQUIRED"
    )
)


# =========================================================
# OUTPUT
# =========================================================

print()

print(
    "=== GOLDEN V2 QA GENERATION V3 768-TOKEN CANARY RESULT ==="
)

print()

print(
    "Results reconstructed:",
    len(
        cell35_canary_results
    ),
)

print(
    "Exact valid checkpoints:",
    cell35_final_valid_checkpoint_count,
)

print(
    "Missing checkpoints:",
    cell35_final_missing_checkpoint_count,
)

print(
    "Invalid checkpoints:",
    cell35_final_invalid_checkpoint_count,
)

print(
    "Resolution modes:",
    dict(
        Counter(
            row[
                "resolution_mode"
            ]
            for row in cell35_resolution_modes
        )
    ),
)

print(
    "Checkpoints created this execution:",
    cell35_checkpoints_created_this_execution,
)

print(
    "API calls this execution:",
    cell35_api_calls_this_execution,
)

print()

print(
    "Structural PASS:",
    cell35_structural_pass_count,
    "/ 12",
)

print(
    "API failures:",
    cell35_api_failure_count,
)

print(
    "Finish reason distribution:",
    dict(
        cell35_finish_reason_distribution
    ),
)

print()

print(
    "Input tokens:",
    cell35_total_input_tokens,
)

print(
    "Output tokens:",
    cell35_total_output_tokens,
)

print(
    "Total tokens:",
    cell35_total_tokens,
)

print(
    "Reasoning tokens:",
    cell35_total_reasoning_tokens,
)

print(
    "Estimated total cost (toman):",
    cell35_total_estimated_cost_toman,
)

print()

print(
    "Runner checks:"
)

for (
    check_name,
    check_value,
) in cell35_runner_checks.items():

    print(
        " -",
        check_name,
        ":",
        (
            "PASS"
            if check_value
            else "FAIL"
        ),
    )

print()

print(
    "Runner health:",
    cell35_runner_status,
)

print(
    "Experimental structural result:",
    cell35_experimental_status,
)

print(
    "Full Generation started:",
    False,
)

print(
    "Next readiness:",
    cell35_next_readiness,
)

print(
    "CELL 35 CHECKPOINTED RUNNER:",
    cell35_runner_status,
)

assert cell35_runner_status == "PASS", (
    "Cell 35 checkpointed runner health failed."
)
````

اجرای ۷۶۸ توکنی Generation v3 برای ۱۱ مورد از ۱۲ نمونه Canary با موفقیت کامل شد و تنها یک مورد به‌دلیل رسیدن به محدودیت طول خروجی متوقف شد.
در این مرحله فقط همان مورد ناقص با سقف خروجی بزرگ‌تر دوباره اجرا می‌شود و ۱۱ نتیجه موفق قبلی بدون اجرای مجدد حفظ می‌شوند.
نتیجه جدید دارای Fingerprint و Checkpoint مستقل است و هیچ Full Generation در این مرحله انجام نمی‌شود.

````python
# Code Cell 36 — Recover Single Truncated QA Canary

from pathlib import Path
import hashlib
import json
import os
import re
import tempfile
import time


# =========================================================
# CONFIG
# =========================================================

CELL36_RUNNER_VERSION = (
    "golden_v2_generation_canary_v3_single_recovery_v1"
)

CELL36_PROMPT_VERSION = (
    "golden_v2_qa_generation_prompt_v3"
)

CELL36_OUTPUT_SCHEMA_VERSION = (
    "golden_v2_qa_generation_output_v3"
)

CELL36_MAX_TOKENS = 1536
CELL36_TEMPERATURE = 0.7
CELL36_TOP_P = 0.8

CELL36_MAX_ATTEMPTS = 2
CELL36_RETRY_SECONDS = 2.0

# Intentional paid recovery of ONLY the truncated Canary.
# MUST become False before final Restart -> Run All.
CELL36_ALLOW_API_FOR_MISSING = False


# =========================================================
# REQUIRED STATE
# =========================================================

required_cell36_names = [
    "project_root",
    "semantic_review_client",
    "semantic_review_model",
    "cell35_canary_results",
    "cell35_canary_requests",
    "cell35_skeleton_by_candidate_id",
    "GENERATION_V3_VERSION",
    "GENERATION_V3_SYSTEM_PROMPT",
    "GENERATION_V3_PROTOCOL_SHA256",
]

missing_cell36_names = [
    name
    for name in required_cell36_names
    if name not in globals()
]

assert not missing_cell36_names, (
    "Missing required Cell 36 state: "
    + ", ".join(missing_cell36_names)
)

assert len(cell35_canary_results) == 12
assert len(cell35_canary_requests) == 12


# =========================================================
# IDENTIFY EXACTLY ONE TRUNCATED RESULT
# =========================================================

cell36_failed_results = [
    result
    for result in cell35_canary_results
    if result.get("structural_status") != "PASS"
]

assert len(cell36_failed_results) == 1, (
    "Expected exactly one failed Cell 35 Canary, got "
    f"{len(cell36_failed_results)}."
)

cell36_failed_result = cell36_failed_results[0]

cell36_candidate_id = cell36_failed_result[
    "candidate_id"
]

assert (
    cell36_failed_result.get("finish_reason")
    == "length"
), (
    "The single failed Canary is not a length truncation."
)

assert (
    "JSON_PARSE_FAILED"
    in (
        cell36_failed_result.get(
            "structural_errors"
        )
        or []
    )
), (
    "The single failed Canary does not have the expected "
    "JSON_PARSE_FAILED truncation signature."
)

assert (
    cell36_candidate_id
    in cell35_skeleton_by_candidate_id
)

cell36_skeleton = (
    cell35_skeleton_by_candidate_id[
        cell36_candidate_id
    ]
)

cell36_original_request = next(
    request
    for request in cell35_canary_requests
    if request["candidate_id"]
    == cell36_candidate_id
)


# =========================================================
# HASH UTILITIES
# =========================================================

def cell36_canonical_json(value):

    return json.dumps(
        value,
        ensure_ascii=False,
        sort_keys=True,
        separators=(",", ":"),
    )


def cell36_sha256_text(text):

    return hashlib.sha256(
        text.encode("utf-8")
    ).hexdigest()


# =========================================================
# OUTPUT VALIDATION
# =========================================================

def cell36_validate_output(
    parsed_json,
):

    errors = []

    if not isinstance(parsed_json, dict):
        return ["OUTPUT_NOT_OBJECT"]

    expected_fields = {
        "question",
        "expected_answer",
        "question_intent",
    }

    if set(parsed_json.keys()) != expected_fields:
        errors.append(
            "OUTPUT_FIELDS_MISMATCH"
        )

    for field_name in expected_fields:

        value = parsed_json.get(
            field_name
        )

        if (
            not isinstance(value, str)
            or not value.strip()
        ):
            errors.append(
                f"INVALID_{field_name.upper()}"
            )

    question = parsed_json.get(
        "question"
    )

    if isinstance(question, str):

        forbidden_question_phrases = [
            "طبق متن بالا",
            "در متن بالا",
            "طبق شواهد",
            "بر اساس شواهد",
            "طبق claim",
            "طبق evidence",
            "در dataset",
        ]

        question_lower = question.lower()

        if any(
            phrase.lower() in question_lower
            for phrase in forbidden_question_phrases
        ):
            errors.append(
                "QUESTION_CONTAINS_CONTEXT_REFERENCE"
            )

    return sorted(
        set(errors)
    )


# =========================================================
# RESPONSE UTILITIES
# =========================================================

def cell36_clean_model_json(
    content,
):

    if not isinstance(content, str):
        return ""

    clean = content.strip()

    if clean.startswith("```"):

        clean = re.sub(
            r"^```(?:json)?\s*",
            "",
            clean,
            flags=re.IGNORECASE,
        )

        clean = re.sub(
            r"\s*```$",
            "",
            clean,
        )

    return clean.strip()


def cell36_is_rate_limit_exception(
    exc,
):

    error_text = str(exc).lower()
    error_type = type(exc).__name__.lower()

    return (
        "ratelimit" in error_type
        or "rate_limit" in error_type
        or "429" in error_text
        or "too many requests" in error_text
    )


# =========================================================
# EXACT REQUEST FINGERPRINT
# =========================================================

cell36_user_prompt = (
    cell36_original_request[
        "user_prompt"
    ]
)

cell36_fingerprint_payload = {
    "runner_version": (
        CELL36_RUNNER_VERSION
    ),

    "candidate_id": (
        cell36_candidate_id
    ),

    "model": (
        semantic_review_model
    ),

    "protocol_version": (
        GENERATION_V3_VERSION
    ),

    "protocol_sha256": (
        GENERATION_V3_PROTOCOL_SHA256
    ),

    "prompt_version": (
        CELL36_PROMPT_VERSION
    ),

    "output_schema_version": (
        CELL36_OUTPUT_SCHEMA_VERSION
    ),

    "system_prompt": (
        GENERATION_V3_SYSTEM_PROMPT
    ),

    "user_prompt": (
        cell36_user_prompt
    ),

    "temperature": (
        CELL36_TEMPERATURE
    ),

    "top_p": (
        CELL36_TOP_P
    ),

    "max_tokens": (
        CELL36_MAX_TOKENS
    ),

    "enable_thinking": False,
}

cell36_request_fingerprint = (
    cell36_sha256_text(
        cell36_canonical_json(
            cell36_fingerprint_payload
        )
    )
)

cell36_request = {
    "candidate_id": (
        cell36_candidate_id
    ),

    "request_fingerprint": (
        cell36_request_fingerprint
    ),

    "user_prompt": (
        cell36_user_prompt
    ),
}


# =========================================================
# CHECKPOINT
# =========================================================

cell36_checkpoint_dir = (
    Path(project_root)
    / "artifacts"
    / "phase2"
    / "golden_dataset_v2"
    / "work"
    / "generation_canary_v3_single_recovery_1536"
)

cell36_checkpoint_dir.mkdir(
    parents=True,
    exist_ok=True,
)


def cell36_checkpoint_path():

    safe_id = re.sub(
        r"[^A-Za-z0-9_.-]+",
        "_",
        cell36_candidate_id,
    )

    return (
        cell36_checkpoint_dir
        / f"{safe_id}.json"
    )


def cell36_atomic_write_json(
    path,
    payload,
):

    path = Path(path)

    content = (
        json.dumps(
            payload,
            ensure_ascii=False,
            sort_keys=True,
            indent=2,
        )
        + "\n"
    )

    fd, temp_name = tempfile.mkstemp(
        prefix=path.name + ".",
        suffix=".tmp",
        dir=str(path.parent),
    )

    try:

        with os.fdopen(
            fd,
            "w",
            encoding="utf-8",
        ) as handle:

            handle.write(content)
            handle.flush()
            os.fsync(handle.fileno())

        os.replace(
            temp_name,
            path,
        )

    except Exception:

        try:
            os.unlink(temp_name)
        except FileNotFoundError:
            pass

        raise


def cell36_build_checkpoint(
    result,
):

    checkpoint = {
        "checkpoint_version": (
            CELL36_RUNNER_VERSION
        ),

        "candidate_id": (
            cell36_candidate_id
        ),

        "request_fingerprint": (
            cell36_request_fingerprint
        ),

        "protocol_version": (
            GENERATION_V3_VERSION
        ),

        "protocol_sha256": (
            GENERATION_V3_PROTOCOL_SHA256
        ),

        "prompt_version": (
            CELL36_PROMPT_VERSION
        ),

        "output_schema_version": (
            CELL36_OUTPUT_SCHEMA_VERSION
        ),

        "model": (
            semantic_review_model
        ),

        "max_tokens": (
            CELL36_MAX_TOKENS
        ),

        "result": result,
    }

    checkpoint[
        "checkpoint_payload_sha256"
    ] = cell36_sha256_text(
        cell36_canonical_json(
            checkpoint
        )
    )

    return checkpoint


def cell36_validate_checkpoint(
    checkpoint,
):

    errors = []

    if not isinstance(checkpoint, dict):
        return ["CHECKPOINT_NOT_OBJECT"]

    stored_sha = checkpoint.get(
        "checkpoint_payload_sha256"
    )

    payload_without_sha = {
        key: value
        for key, value in checkpoint.items()
        if key != "checkpoint_payload_sha256"
    }

    recalculated_sha = (
        cell36_sha256_text(
            cell36_canonical_json(
                payload_without_sha
            )
        )
    )

    if stored_sha != recalculated_sha:
        errors.append(
            "CHECKPOINT_HASH_MISMATCH"
        )

    exact_expectations = {
        "checkpoint_version": (
            CELL36_RUNNER_VERSION
        ),
        "candidate_id": (
            cell36_candidate_id
        ),
        "request_fingerprint": (
            cell36_request_fingerprint
        ),
        "protocol_version": (
            GENERATION_V3_VERSION
        ),
        "protocol_sha256": (
            GENERATION_V3_PROTOCOL_SHA256
        ),
        "prompt_version": (
            CELL36_PROMPT_VERSION
        ),
        "output_schema_version": (
            CELL36_OUTPUT_SCHEMA_VERSION
        ),
        "model": (
            semantic_review_model
        ),
        "max_tokens": (
            CELL36_MAX_TOKENS
        ),
    }

    for (
        field_name,
        expected_value,
    ) in exact_expectations.items():

        if (
            checkpoint.get(field_name)
            != expected_value
        ):
            errors.append(
                f"{field_name.upper()}_MISMATCH"
            )

    result = checkpoint.get(
        "result"
    )

    if not isinstance(result, dict):

        errors.append(
            "CHECKPOINT_RESULT_INVALID"
        )

    elif (
        result.get("candidate_id")
        != cell36_candidate_id
    ):

        errors.append(
            "CHECKPOINT_RESULT_CANDIDATE_MISMATCH"
        )

    return errors


def cell36_load_checkpoint():

    path = cell36_checkpoint_path()

    if not path.exists():

        return (
            None,
            "MISSING",
            [],
        )

    try:

        with path.open(
            "r",
            encoding="utf-8",
        ) as handle:

            checkpoint = json.load(
                handle
            )

    except Exception as exc:

        return (
            None,
            "UNREADABLE",
            [
                f"{type(exc).__name__}: {exc}"
            ],
        )

    errors = (
        cell36_validate_checkpoint(
            checkpoint
        )
    )

    if errors:

        return (
            checkpoint,
            "INVALID",
            errors,
        )

    return (
        checkpoint,
        "VALID",
        [],
    )


# =========================================================
# COST
# =========================================================

cell36_has_cost_rates = (
    "INPUT_TOMAN_PER_TOKEN" in globals()
    and
    "OUTPUT_TOMAN_PER_TOKEN" in globals()
)


def cell36_estimated_cost_toman(
    input_tokens,
    output_tokens,
):

    if not cell36_has_cost_rates:
        return None

    return (
        input_tokens
        * INPUT_TOMAN_PER_TOKEN
        +
        output_tokens
        * OUTPUT_TOMAN_PER_TOKEN
    )


# =========================================================
# API EXECUTION
# =========================================================

def cell36_execute_request():

    attempt_history = []

    for attempt_number in range(
        1,
        CELL36_MAX_ATTEMPTS + 1,
    ):

        started_at = time.time()

        try:

            response = (
                semantic_review_client
                .chat
                .completions
                .create(
                    model=(
                        semantic_review_model
                    ),

                    messages=[
                        {
                            "role": "system",
                            "content": (
                                GENERATION_V3_SYSTEM_PROMPT
                            ),
                        },
                        {
                            "role": "user",
                            "content": (
                                cell36_user_prompt
                            ),
                        },
                    ],

                    temperature=(
                        CELL36_TEMPERATURE
                    ),

                    top_p=(
                        CELL36_TOP_P
                    ),

                    max_tokens=(
                        CELL36_MAX_TOKENS
                    ),

                    extra_body={
                        "enable_thinking": False,
                    },
                )
            )

        except Exception as exc:

            attempt_history.append({
                "attempt": attempt_number,
                "status": "API_ERROR",
                "error": (
                    f"{type(exc).__name__}: {exc}"
                ),
            })

            if (
                attempt_number
                < CELL36_MAX_ATTEMPTS
            ):

                wait_seconds = (
                    70.0
                    if cell36_is_rate_limit_exception(
                        exc
                    )
                    else (
                        CELL36_RETRY_SECONDS
                        * attempt_number
                    )
                )

                time.sleep(
                    wait_seconds
                )

                continue

            return {
                "candidate_id": (
                    cell36_candidate_id
                ),
                "execution_status": (
                    "API_FAILED"
                ),
                "attempt_history": (
                    attempt_history
                ),
            }

        elapsed_seconds = (
            time.time()
            - started_at
        )

        choice = response.choices[0]
        message = choice.message

        finish_reason = getattr(
            choice,
            "finish_reason",
            None,
        )

        raw_content = getattr(
            message,
            "content",
            None,
        )

        usage = getattr(
            response,
            "usage",
            None,
        )

        input_tokens = int(
            getattr(
                usage,
                "prompt_tokens",
                0,
            )
            or 0
        )

        output_tokens = int(
            getattr(
                usage,
                "completion_tokens",
                0,
            )
            or 0
        )

        total_tokens = int(
            getattr(
                usage,
                "total_tokens",
                0,
            )
            or (
                input_tokens
                + output_tokens
            )
        )

        reasoning_tokens = 0

        completion_details = getattr(
            usage,
            "completion_tokens_details",
            None,
        )

        if completion_details is not None:

            reasoning_tokens = int(
                getattr(
                    completion_details,
                    "reasoning_tokens",
                    0,
                )
                or 0
            )

        clean_content = (
            cell36_clean_model_json(
                raw_content
            )
        )

        parsed_json = None
        parse_error = None

        try:

            parsed_json = json.loads(
                clean_content
            )

        except Exception as exc:

            parse_error = (
                f"{type(exc).__name__}: {exc}"
            )

        if parsed_json is None:

            structural_errors = [
                "JSON_PARSE_FAILED"
            ]

        else:

            structural_errors = (
                cell36_validate_output(
                    parsed_json
                )
            )

        return {
            "candidate_id": (
                cell36_candidate_id
            ),

            "request_fingerprint": (
                cell36_request_fingerprint
            ),

            "model": (
                semantic_review_model
            ),

            "protocol_version": (
                GENERATION_V3_VERSION
            ),

            "protocol_sha256": (
                GENERATION_V3_PROTOCOL_SHA256
            ),

            "prompt_version": (
                CELL36_PROMPT_VERSION
            ),

            "output_schema_version": (
                CELL36_OUTPUT_SCHEMA_VERSION
            ),

            "max_tokens": (
                CELL36_MAX_TOKENS
            ),

            "attempt_number": (
                attempt_number
            ),

            "finish_reason": (
                finish_reason
            ),

            "elapsed_seconds": round(
                elapsed_seconds,
                4,
            ),

            "usage": {
                "input_tokens": input_tokens,
                "output_tokens": output_tokens,
                "total_tokens": total_tokens,
                "reasoning_tokens": reasoning_tokens,
            },

            "estimated_cost_toman": (
                cell36_estimated_cost_toman(
                    input_tokens,
                    output_tokens,
                )
            ),

            "raw_content": raw_content,

            "parsed_json": parsed_json,

            "parse_error": parse_error,

            "structural_errors": (
                structural_errors
            ),

            "structural_status": (
                "PASS"
                if not structural_errors
                else "FAIL"
            ),

            "attempt_history": (
                attempt_history
            ),
        }

    raise RuntimeError(
        "Unexpected Cell 36 execution path."
    )


# =========================================================
# PREFLIGHT
# =========================================================

print(
    "=== GOLDEN V2 SINGLE CANARY RECOVERY PREFLIGHT ==="
)

print()

print(
    "Recovery population:",
    1,
)

print(
    "Candidate:",
    cell36_candidate_id,
)

print(
    "Previous finish reason:",
    cell36_failed_result.get(
        "finish_reason"
    ),
)

print(
    "Previous structural errors:",
    cell36_failed_result.get(
        "structural_errors"
    ),
)

print(
    "Model:",
    semantic_review_model,
)

print(
    "Protocol:",
    GENERATION_V3_VERSION,
)

print(
    "Max tokens:",
    CELL36_MAX_TOKENS,
)

print(
    "API allowed for missing exact checkpoint:",
    CELL36_ALLOW_API_FOR_MISSING,
)

print(
    "Full Generation enabled:",
    False,
)

print()


# =========================================================
# RESOLVE SINGLE RECOVERY
# =========================================================

(
    cell36_checkpoint,
    cell36_checkpoint_status,
    cell36_checkpoint_errors,
) = cell36_load_checkpoint()

cell36_api_calls_this_execution = 0
cell36_checkpoint_created_this_execution = 0


if cell36_checkpoint_status == "VALID":

    cell36_recovery_result = (
        cell36_checkpoint[
            "result"
        ]
    )

    cell36_resolution_mode = (
        "EXACT_CHECKPOINT"
    )

else:

    if cell36_checkpoint_status in {
        "INVALID",
        "UNREADABLE",
    }:

        raise AssertionError(
            "Existing Cell 36 checkpoint is invalid. "
            "Refusing automatic paid replacement. "
            f"Status={cell36_checkpoint_status}; "
            f"errors={cell36_checkpoint_errors}"
        )

    if not CELL36_ALLOW_API_FOR_MISSING:

        raise AssertionError(
            "Exact Cell 36 checkpoint is missing and "
            "API execution is disabled."
        )

    cell36_recovery_result = (
        cell36_execute_request()
    )

    cell36_api_calls_this_execution = 1

    cell36_checkpoint = (
        cell36_build_checkpoint(
            cell36_recovery_result
        )
    )

    cell36_atomic_write_json(
        cell36_checkpoint_path(),
        cell36_checkpoint,
    )

    (
        verified_checkpoint,
        verified_status,
        verified_errors,
    ) = cell36_load_checkpoint()

    assert verified_status == "VALID", (
        "Cell 36 checkpoint verification failed: "
        f"{verified_errors}"
    )

    cell36_recovery_result = (
        verified_checkpoint[
            "result"
        ]
    )

    cell36_resolution_mode = (
        "NEW_PAID_EXECUTION"
    )

    cell36_checkpoint_created_this_execution = 1


# =========================================================
# BUILD FINAL 12-CANARY RESULT SET
# =========================================================

cell36_final_canary_results = []

for result in cell35_canary_results:

    if (
        result["candidate_id"]
        == cell36_candidate_id
    ):

        cell36_final_canary_results.append(
            cell36_recovery_result
        )

    else:

        cell36_final_canary_results.append(
            result
        )


assert len(cell36_final_canary_results) == 12
assert len({
    result["candidate_id"]
    for result in cell36_final_canary_results
}) == 12


# =========================================================
# FINAL STATUS
# =========================================================

cell36_structural_pass_count = sum(
    result.get(
        "structural_status"
    )
    == "PASS"
    for result in cell36_final_canary_results
)

cell36_recovery_status = (
    cell36_recovery_result.get(
        "structural_status"
    )
)

cell36_api_failure = (
    cell36_recovery_result.get(
        "execution_status"
    )
    == "API_FAILED"
)

(
    final_checkpoint,
    final_checkpoint_status,
    final_checkpoint_errors,
) = cell36_load_checkpoint()


cell36_runner_checks = {
    "exactly_one_recovery_target": (
        len(cell36_failed_results) == 1
    ),

    "recovery_checkpoint_valid": (
        final_checkpoint_status
        == "VALID"
    ),

    "zero_api_failure": (
        not cell36_api_failure
    ),

    "final_canary_population_12": (
        len(cell36_final_canary_results)
        == 12
    ),

    "unique_final_candidates_12": (
        len({
            result["candidate_id"]
            for result
            in cell36_final_canary_results
        })
        == 12
    ),

    "full_generation_not_started": True,
}

cell36_runner_status = (
    "PASS"
    if all(
        cell36_runner_checks.values()
    )
    else "FAIL"
)

cell36_experimental_status = (
    "PASS"
    if cell36_structural_pass_count
    == 12
    else "FAIL"
)

cell36_next_readiness = (
    "READY_FOR_QA_GENERATION_V3_SEMANTIC_CANARY_VALIDATION"
    if (
        cell36_runner_status == "PASS"
        and cell36_experimental_status == "PASS"
    )
    else (
        "READY_FOR_CELL36_FAILURE_ANALYSIS"
        if cell36_runner_status == "PASS"
        else "BLOCKED_CELL36_RUNNER_REPAIR_REQUIRED"
    )
)


# =========================================================
# OUTPUT
# =========================================================

print(
    "=== GOLDEN V2 SINGLE CANARY RECOVERY RESULT ==="
)

print()

print(
    "Candidate:",
    cell36_candidate_id,
)

print(
    "Resolution mode:",
    cell36_resolution_mode,
)

print(
    "API calls this execution:",
    cell36_api_calls_this_execution,
)

print(
    "Checkpoint created this execution:",
    cell36_checkpoint_created_this_execution,
)

print()

print(
    "Recovery finish reason:",
    cell36_recovery_result.get(
        "finish_reason"
    ),
)

print(
    "Recovery structural status:",
    cell36_recovery_status,
)

print(
    "Recovery structural errors:",
    cell36_recovery_result.get(
        "structural_errors"
    ),
)

print(
    "Recovery parse error:",
    cell36_recovery_result.get(
        "parse_error"
    ),
)

print()

print(
    "Recovery usage:",
    cell36_recovery_result.get(
        "usage"
    ),
)

print(
    "Recovery estimated cost (toman):",
    cell36_recovery_result.get(
        "estimated_cost_toman"
    ),
)

print()

print(
    "Final Canary Structural PASS:",
    cell36_structural_pass_count,
    "/ 12",
)

print()

print(
    "Runner checks:"
)

for (
    check_name,
    check_value,
) in cell36_runner_checks.items():

    print(
        " -",
        check_name,
        ":",
        (
            "PASS"
            if check_value
            else "FAIL"
        ),
    )

print()

print(
    "Runner health:",
    cell36_runner_status,
)

print(
    "Experimental structural result:",
    cell36_experimental_status,
)

print(
    "Full Generation started:",
    False,
)

print(
    "Next readiness:",
    cell36_next_readiness,
)

print(
    "CELL 36 CHECKPOINTED RUNNER:",
    cell36_runner_status,
)

assert cell36_runner_status == "PASS", (
    "Cell 36 checkpointed runner health failed."
)
````

<div style="direction: rtl; white-space: normal; line-height: 1.7;">
<h2>مرحله ۳۷ — اعتبارسنجی معنایی Canary سؤال و پاسخ</h2>
<p>
در این مرحله ۱۲ خروجی نهایی Canary از نظر معنایی در برابر Claim هدف و Evidence بررسی می‌شوند.
اعتبار Claim و ارتباط Claim با Evidence قبلاً در NB12i تأیید شده و در این مرحله دوباره داوری نمی‌شود.
تمرکز اعتبارسنجی فقط بر کیفیت Question و Expected Answer است: سؤال باید مستقیماً Claim را هدف بگیرد، پاسخ باید معنای Claim را با حفظ دامنه و قیود بازتاب دهد، و هیچ انحراف به اطلاعات جانبی یا دانش خارجی رخ ندهد.
نتایج هر داوری با Fingerprint دقیق و Checkpoint مستقل ذخیره می‌شوند و Full Generation در این مرحله آغاز نمی‌شود.
</p>
</div>

````python
# Code Cell 37 — QA Generation v3 Semantic Canary Validation

from collections import Counter
from pathlib import Path
import hashlib
import json
import os
import re
import tempfile
import time


# =========================================================
# CONFIG
# =========================================================

CELL37_RUNNER_VERSION = (
    "golden_v2_qa_semantic_canary_validator_v1"
)

CELL37_VALIDATION_PROTOCOL_VERSION = (
    "golden_v2_qa_semantic_validation_v1"
)

CELL37_OUTPUT_SCHEMA_VERSION = (
    "golden_v2_qa_semantic_validation_output_v1"
)

CELL37_MAX_TOKENS = 768
CELL37_TEMPERATURE = 0.0
CELL37_TOP_P = 1.0

CELL37_MAX_ATTEMPTS = 2
CELL37_RETRY_SECONDS = 2.0
CELL37_REQUEST_SPACING_SECONDS = 2.7

# Historical paid Canary is complete.
# Restart -> Run All MUST NOT create new paid requests.
CELL37_ALLOW_API_FOR_MISSING = False


# =========================================================
# REQUIRED STATE
# =========================================================

required_cell37_names = [
    "project_root",
    "semantic_review_client",
    "semantic_review_model",
    "cell25_candidate_skeletons",
    "cell36_final_canary_results",
    "GENERATION_V3_VERSION",
    "GENERATION_V3_PROTOCOL_SHA256",
]

missing_cell37_names = [
    name
    for name in required_cell37_names
    if name not in globals()
]

assert not missing_cell37_names, (
    "Missing required Cell 37 state: "
    + ", ".join(missing_cell37_names)
)

assert len(cell25_candidate_skeletons) == 3813
assert len(cell36_final_canary_results) == 12

assert len({
    result["candidate_id"]
    for result in cell36_final_canary_results
}) == 12

assert all(
    result.get("structural_status") == "PASS"
    for result in cell36_final_canary_results
), (
    "Cell 37 requires 12/12 structurally valid generated Canary outputs."
)


# =========================================================
# LOOKUPS
# =========================================================

cell37_skeleton_by_candidate_id = {
    record["candidate_id"]: record
    for record in cell25_candidate_skeletons
}

assert len(cell37_skeleton_by_candidate_id) == 3813


cell37_generation_by_candidate_id = {
    result["candidate_id"]: result
    for result in cell36_final_canary_results
}

assert len(cell37_generation_by_candidate_id) == 12


# =========================================================
# HASH UTILITIES
# =========================================================

def cell37_canonical_json(value):

    return json.dumps(
        value,
        ensure_ascii=False,
        sort_keys=True,
        separators=(",", ":"),
    )


def cell37_sha256_text(text):

    return hashlib.sha256(
        text.encode("utf-8")
    ).hexdigest()


# =========================================================
# SEMANTIC VALIDATION CONTRACT
# =========================================================

CELL37_SYSTEM_PROMPT = """
You are validating a generated Question and Expected Answer for a
pre-validated Golden Dataset candidate.

Architecture rule:
The target Claim and its Evidence were already semantically validated
upstream. Do NOT decide whether the Claim itself is true or whether the
Evidence supports the Claim. Your task is ONLY to evaluate the generated
Question and Expected Answer against that already validated target.

Evaluate these dimensions:

1. question_targets_claim
   The Question directly asks for the information expressed by the target
   Claim rather than a neighboring or broader fact.

2. answer_recovers_claim
   The Expected Answer substantially preserves the meaning of the target
   Claim.

3. scope_fidelity
   The Question and Expected Answer preserve relevant scope, conditions,
   population, comparison, time, or domain restrictions.

4. modality_quantifier_fidelity
   Relevant modality and quantifiers such as may, must, generally, many,
   only, percentages, comparisons, or conditions are not strengthened,
   weakened, dropped, or invented.

5. standalone_question
   The Question is understandable without access to hidden context and does
   not refer to "the text above", "the evidence", "the claim", or the dataset.

6. no_neighbor_fact_drift
   The Question and Expected Answer do not replace the target Claim with
   another assertion merely because it appears in the Evidence.

7. no_external_knowledge
   The Expected Answer does not add substantive information that is absent
   from the target Claim and supplied Evidence.

8. answer_question_alignment
   The Expected Answer actually answers the generated Question.

Verdict rules:

PASS:
All eight dimensions pass.

FAIL:
One or more dimensions materially fail.

Return JSON only with exactly these fields:

{
  "verdict": "PASS",
  "question_targets_claim": true,
  "answer_recovers_claim": true,
  "scope_fidelity": true,
  "modality_quantifier_fidelity": true,
  "standalone_question": true,
  "no_neighbor_fact_drift": true,
  "no_external_knowledge": true,
  "answer_question_alignment": true,
  "failure_reasons": []
}

If verdict is FAIL, failure_reasons must contain concise machine-readable
reason strings describing the failed dimensions.

Do not output markdown.
Do not output explanations outside the JSON.
""".strip()


CELL37_BOOLEAN_FIELDS = [
    "question_targets_claim",
    "answer_recovers_claim",
    "scope_fidelity",
    "modality_quantifier_fidelity",
    "standalone_question",
    "no_neighbor_fact_drift",
    "no_external_knowledge",
    "answer_question_alignment",
]

CELL37_EXPECTED_FIELDS = {
    "verdict",
    *CELL37_BOOLEAN_FIELDS,
    "failure_reasons",
}


# =========================================================
# INPUT PAYLOAD
# =========================================================

def cell37_build_payload(candidate_id):

    skeleton = (
        cell37_skeleton_by_candidate_id[
            candidate_id
        ]
    )

    generation_result = (
        cell37_generation_by_candidate_id[
            candidate_id
        ]
    )

    generated = generation_result.get(
        "parsed_json"
    )

    assert isinstance(generated, dict)

    evidence = []

    for (
        evidence_id,
        evidence_quote,
        evidence_locator,
    ) in zip(
        skeleton["evidence_ids"],
        skeleton["evidence_quotes"],
        skeleton["evidence_locators"],
    ):

        evidence.append({
            "evidence_id": evidence_id,
            "quote": evidence_quote,
            "locator": evidence_locator,
        })

    return {
        "candidate_id": candidate_id,

        "target_claim": {
            "claim_id": skeleton["claim_id"],
            "text_fa": skeleton["claim_text_fa"],
            "claim_type": skeleton["claim_type"],
            "claim_types": skeleton["claim_types"],
            "authority_class": skeleton["authority_class"],
        },

        "evidence": evidence,

        "generated_qa": {
            "question": generated["question"],
            "expected_answer": generated["expected_answer"],
            "question_intent": generated["question_intent"],
        },
    }


def cell37_build_user_prompt(candidate_id):

    payload = cell37_build_payload(
        candidate_id
    )

    return (
        "Validate only the generated Question and Expected Answer "
        "against the pre-validated target Claim and Evidence.\n\n"
        + json.dumps(
            payload,
            ensure_ascii=False,
            sort_keys=True,
            indent=2,
        )
    )


# =========================================================
# OUTPUT VALIDATION
# =========================================================

def cell37_validate_output(parsed_json):

    errors = []

    if not isinstance(parsed_json, dict):
        return ["OUTPUT_NOT_OBJECT"]

    if set(parsed_json.keys()) != CELL37_EXPECTED_FIELDS:
        errors.append(
            "OUTPUT_FIELDS_MISMATCH"
        )

    verdict = parsed_json.get(
        "verdict"
    )

    if verdict not in {
        "PASS",
        "FAIL",
    }:
        errors.append(
            "INVALID_VERDICT"
        )

    for field_name in CELL37_BOOLEAN_FIELDS:

        if not isinstance(
            parsed_json.get(field_name),
            bool,
        ):
            errors.append(
                f"INVALID_{field_name.upper()}"
            )

    failure_reasons = parsed_json.get(
        "failure_reasons"
    )

    if not isinstance(
        failure_reasons,
        list,
    ):
        errors.append(
            "INVALID_FAILURE_REASONS"
        )

    elif not all(
        isinstance(item, str)
        and item.strip()
        for item in failure_reasons
    ):
        errors.append(
            "INVALID_FAILURE_REASON_ITEM"
        )

    if (
        verdict == "PASS"
        and isinstance(
            failure_reasons,
            list,
        )
        and failure_reasons
    ):
        errors.append(
            "PASS_WITH_FAILURE_REASONS"
        )

    if (
        verdict == "PASS"
        and any(
            parsed_json.get(field_name)
            is not True
            for field_name
            in CELL37_BOOLEAN_FIELDS
        )
    ):
        errors.append(
            "PASS_WITH_FAILED_DIMENSION"
        )

    if (
        verdict == "FAIL"
        and all(
            parsed_json.get(field_name)
            is True
            for field_name
            in CELL37_BOOLEAN_FIELDS
        )
    ):
        errors.append(
            "FAIL_WITHOUT_FAILED_DIMENSION"
        )

    if (
        verdict == "FAIL"
        and isinstance(
            failure_reasons,
            list,
        )
        and not failure_reasons
    ):
        errors.append(
            "FAIL_WITHOUT_FAILURE_REASON"
        )

    return sorted(
        set(errors)
    )


# =========================================================
# RESPONSE CLEANING
# =========================================================

def cell37_clean_model_json(content):

    if not isinstance(content, str):
        return ""

    clean = content.strip()

    if clean.startswith("```"):

        clean = re.sub(
            r"^```(?:json)?\s*",
            "",
            clean,
            flags=re.IGNORECASE,
        )

        clean = re.sub(
            r"\s*```$",
            "",
            clean,
        )

    return clean.strip()


def cell37_is_rate_limit_exception(exc):

    error_text = str(exc).lower()
    error_type = type(exc).__name__.lower()

    return (
        "ratelimit" in error_type
        or "rate_limit" in error_type
        or "429" in error_text
        or "too many requests" in error_text
    )


# =========================================================
# REQUEST CONTRACT
# =========================================================

def cell37_build_request(candidate_id):

    user_prompt = (
        cell37_build_user_prompt(
            candidate_id
        )
    )

    generation_result = (
        cell37_generation_by_candidate_id[
            candidate_id
        ]
    )

    generated_qa_sha256 = (
        cell37_sha256_text(
            cell37_canonical_json(
                generation_result[
                    "parsed_json"
                ]
            )
        )
    )

    fingerprint_payload = {
        "runner_version": (
            CELL37_RUNNER_VERSION
        ),

        "validation_protocol_version": (
            CELL37_VALIDATION_PROTOCOL_VERSION
        ),

        "output_schema_version": (
            CELL37_OUTPUT_SCHEMA_VERSION
        ),

        "candidate_id": candidate_id,

        "generation_protocol_version": (
            GENERATION_V3_VERSION
        ),

        "generation_protocol_sha256": (
            GENERATION_V3_PROTOCOL_SHA256
        ),

        "generated_qa_sha256": (
            generated_qa_sha256
        ),

        "model": (
            semantic_review_model
        ),

        "system_prompt": (
            CELL37_SYSTEM_PROMPT
        ),

        "user_prompt": (
            user_prompt
        ),

        "temperature": (
            CELL37_TEMPERATURE
        ),

        "top_p": (
            CELL37_TOP_P
        ),

        "max_tokens": (
            CELL37_MAX_TOKENS
        ),

        "enable_thinking": False,
    }

    request_fingerprint = (
        cell37_sha256_text(
            cell37_canonical_json(
                fingerprint_payload
            )
        )
    )

    return {
        "candidate_id": candidate_id,

        "user_prompt": user_prompt,

        "generated_qa_sha256": (
            generated_qa_sha256
        ),

        "request_fingerprint": (
            request_fingerprint
        ),
    }


cell37_canary_candidate_ids = [
    result["candidate_id"]
    for result in cell36_final_canary_results
]

cell37_requests = [
    cell37_build_request(
        candidate_id
    )
    for candidate_id
    in cell37_canary_candidate_ids
]

assert len(cell37_requests) == 12


# =========================================================
# CHECKPOINT DIRECTORY
# =========================================================

cell37_checkpoint_dir = (
    Path(project_root)
    / "artifacts"
    / "phase2"
    / "golden_dataset_v2"
    / "work"
    / "qa_semantic_canary_validation_v1"
)

# Historical directory must already exist.
# This cell does not create missing paid-run infrastructure.
assert cell37_checkpoint_dir.exists(), (
    "Cell 37 historical checkpoint directory is missing: "
    f"{cell37_checkpoint_dir}"
)


def cell37_checkpoint_path(candidate_id):

    safe_id = re.sub(
        r"[^A-Za-z0-9_.-]+",
        "_",
        candidate_id,
    )

    return (
        cell37_checkpoint_dir
        / f"{safe_id}.json"
    )


# =========================================================
# ATOMIC WRITE HELPER
# Retained only for contract compatibility.
# Historical reconstruction path never calls it.
# =========================================================

def cell37_atomic_write_json(
    path,
    payload,
):

    if not CELL37_ALLOW_API_FOR_MISSING:
        raise RuntimeError(
            "Cell 37 is frozen in reconstruction-only mode; "
            "file writes are disabled."
        )

    path = Path(path)

    content = (
        json.dumps(
            payload,
            ensure_ascii=False,
            sort_keys=True,
            indent=2,
        )
        + "\n"
    )

    fd, temp_name = tempfile.mkstemp(
        prefix=path.name + ".",
        suffix=".tmp",
        dir=str(path.parent),
    )

    try:

        with os.fdopen(
            fd,
            "w",
            encoding="utf-8",
        ) as handle:

            handle.write(content)
            handle.flush()
            os.fsync(
                handle.fileno()
            )

        os.replace(
            temp_name,
            path,
        )

    except Exception:

        try:
            os.unlink(
                temp_name
            )
        except FileNotFoundError:
            pass

        raise


# =========================================================
# CHECKPOINT CONTRACT
# =========================================================

def cell37_build_checkpoint(
    request,
    result,
):

    checkpoint = {
        "checkpoint_version": (
            CELL37_RUNNER_VERSION
        ),

        "candidate_id": (
            request[
                "candidate_id"
            ]
        ),

        "request_fingerprint": (
            request[
                "request_fingerprint"
            ]
        ),

        "generated_qa_sha256": (
            request[
                "generated_qa_sha256"
            ]
        ),

        "validation_protocol_version": (
            CELL37_VALIDATION_PROTOCOL_VERSION
        ),

        "output_schema_version": (
            CELL37_OUTPUT_SCHEMA_VERSION
        ),

        "model": (
            semantic_review_model
        ),

        "result": result,
    }

    checkpoint[
        "checkpoint_payload_sha256"
    ] = cell37_sha256_text(
        cell37_canonical_json(
            checkpoint
        )
    )

    return checkpoint


def cell37_validate_checkpoint(
    checkpoint,
    request,
):

    errors = []

    if not isinstance(checkpoint, dict):
        return [
            "CHECKPOINT_NOT_OBJECT"
        ]

    stored_sha = checkpoint.get(
        "checkpoint_payload_sha256"
    )

    payload_without_sha = {
        key: value
        for key, value
        in checkpoint.items()
        if key
        != "checkpoint_payload_sha256"
    }

    recalculated_sha = (
        cell37_sha256_text(
            cell37_canonical_json(
                payload_without_sha
            )
        )
    )

    if stored_sha != recalculated_sha:
        errors.append(
            "CHECKPOINT_HASH_MISMATCH"
        )

    exact_expectations = {
        "checkpoint_version": (
            CELL37_RUNNER_VERSION
        ),

        "candidate_id": (
            request[
                "candidate_id"
            ]
        ),

        "request_fingerprint": (
            request[
                "request_fingerprint"
            ]
        ),

        "generated_qa_sha256": (
            request[
                "generated_qa_sha256"
            ]
        ),

        "validation_protocol_version": (
            CELL37_VALIDATION_PROTOCOL_VERSION
        ),

        "output_schema_version": (
            CELL37_OUTPUT_SCHEMA_VERSION
        ),

        "model": (
            semantic_review_model
        ),
    }

    for (
        field_name,
        expected_value,
    ) in exact_expectations.items():

        if (
            checkpoint.get(
                field_name
            )
            != expected_value
        ):
            errors.append(
                f"{field_name.upper()}_MISMATCH"
            )

    result = checkpoint.get(
        "result"
    )

    if not isinstance(
        result,
        dict,
    ):
        errors.append(
            "CHECKPOINT_RESULT_INVALID"
        )

    elif (
        result.get(
            "candidate_id"
        )
        != request[
            "candidate_id"
        ]
    ):
        errors.append(
            "CHECKPOINT_RESULT_CANDIDATE_MISMATCH"
        )

    return errors


def cell37_load_checkpoint(request):

    path = cell37_checkpoint_path(
        request[
            "candidate_id"
        ]
    )

    if not path.exists():

        return (
            None,
            "MISSING",
            [],
        )

    try:

        with path.open(
            "r",
            encoding="utf-8",
        ) as handle:

            checkpoint = json.load(
                handle
            )

    except Exception as exc:

        return (
            None,
            "UNREADABLE",
            [
                f"{type(exc).__name__}: {exc}"
            ],
        )

    errors = (
        cell37_validate_checkpoint(
            checkpoint,
            request,
        )
    )

    if errors:

        return (
            checkpoint,
            "INVALID",
            errors,
        )

    return (
        checkpoint,
        "VALID",
        [],
    )


# =========================================================
# COST
# =========================================================

cell37_has_cost_rates = (
    "INPUT_TOMAN_PER_TOKEN" in globals()
    and
    "OUTPUT_TOMAN_PER_TOKEN" in globals()
)


def cell37_estimated_cost_toman(
    input_tokens,
    output_tokens,
):

    if not cell37_has_cost_rates:
        return None

    return (
        input_tokens
        * INPUT_TOMAN_PER_TOKEN
        +
        output_tokens
        * OUTPUT_TOMAN_PER_TOKEN
    )


# =========================================================
# API EXECUTION
# Disabled permanently for this historical Cell.
# =========================================================

def cell37_execute_request(request):

    raise RuntimeError(
        "Cell 37 API execution is disabled. "
        "Historical Canary checkpoints must be reconstructed "
        "without new model calls."
    )


# =========================================================
# PREFLIGHT
# =========================================================

print(
    "=== GOLDEN V2 QA SEMANTIC CANARY RECONSTRUCTION PREFLIGHT ==="
)

print()

print(
    "Validation population:",
    len(cell37_requests),
)

print(
    "Model:",
    semantic_review_model,
)

print(
    "Validation protocol:",
    CELL37_VALIDATION_PROTOCOL_VERSION,
)

print(
    "Historical max tokens:",
    CELL37_MAX_TOKENS,
)

print(
    "API allowed for missing exact checkpoints:",
    CELL37_ALLOW_API_FOR_MISSING,
)

print(
    "Mode:",
    "HISTORICAL_CHECKPOINT_RECONSTRUCTION_ONLY",
)

print(
    "Full Generation started:",
    False,
)

print()


# =========================================================
# RECONSTRUCT ONLY
# =========================================================

cell37_results = []
cell37_resolution_modes = []

cell37_api_calls_this_execution = 0
cell37_checkpoints_created_this_execution = 0


for index, request in enumerate(
    cell37_requests,
    start=1,
):

    (
        checkpoint,
        checkpoint_status,
        checkpoint_errors,
    ) = cell37_load_checkpoint(
        request
    )

    candidate_id = request[
        "candidate_id"
    ]

    if checkpoint_status != "VALID":

        raise AssertionError(
            "Cell 37 is reconstruction-only. "
            "An exact historical checkpoint is unavailable or invalid "
            f"for {candidate_id}. "
            f"Status={checkpoint_status}; "
            f"errors={checkpoint_errors}. "
            "No API request was made."
        )

    result = checkpoint[
        "result"
    ]

    resolution_mode = (
        "EXACT_CHECKPOINT"
    )

    cell37_results.append(
        result
    )

    cell37_resolution_modes.append({
        "candidate_id": (
            candidate_id
        ),

        "resolution_mode": (
            resolution_mode
        ),
    })

    print(
        f"[{index:02d}/12]",
        candidate_id,
        "|",
        resolution_mode,
        "| structural=",
        result.get(
            "structural_status",
            result.get(
                "execution_status"
            ),
        ),
        "| semantic=",
        result.get(
            "semantic_verdict"
        ),
        flush=True,
    )


# =========================================================
# FINAL DISK RESCAN
# =========================================================

cell37_valid_checkpoint_count = 0
cell37_missing_checkpoint_count = 0
cell37_invalid_checkpoint_count = 0


for request in cell37_requests:

    (
        checkpoint,
        status,
        errors,
    ) = cell37_load_checkpoint(
        request
    )

    if status == "VALID":
        cell37_valid_checkpoint_count += 1

    elif status == "MISSING":
        cell37_missing_checkpoint_count += 1

    else:
        cell37_invalid_checkpoint_count += 1


# =========================================================
# HISTORICAL RESULT SUMMARY
# =========================================================

cell37_structural_pass_count = sum(
    result.get(
        "structural_status"
    )
    == "PASS"
    for result in cell37_results
)

cell37_api_failure_count = sum(
    result.get(
        "execution_status"
    )
    == "API_FAILED"
    for result in cell37_results
)

cell37_semantic_distribution = Counter(
    result.get(
        "semantic_verdict"
    )
    for result in cell37_results
    if result.get(
        "structural_status"
    )
    == "PASS"
)

cell37_dimension_failures = Counter()

for result in cell37_results:

    parsed = result.get(
        "parsed_json"
    )

    if not isinstance(
        parsed,
        dict,
    ):
        continue

    for field_name in CELL37_BOOLEAN_FIELDS:

        if parsed.get(
            field_name
        ) is False:

            cell37_dimension_failures[
                field_name
            ] += 1


cell37_known_costs = [
    result.get(
        "estimated_cost_toman"
    )
    for result in cell37_results
    if isinstance(
        result.get(
            "estimated_cost_toman"
        ),
        (int, float),
    )
]

cell37_total_estimated_cost_toman = (
    sum(cell37_known_costs)
    if cell37_known_costs
    else None
)


# =========================================================
# RUNNER HEALTH
# Experimental failure is NOT infrastructure failure.
# =========================================================

cell37_runner_checks = {
    "12_results_reconstructed": (
        len(cell37_results)
        == 12
    ),

    "12_valid_checkpoints": (
        cell37_valid_checkpoint_count
        == 12
    ),

    "zero_missing_checkpoints": (
        cell37_missing_checkpoint_count
        == 0
    ),

    "zero_invalid_checkpoints": (
        cell37_invalid_checkpoint_count
        == 0
    ),

    "zero_api_failures": (
        cell37_api_failure_count
        == 0
    ),

    "zero_api_calls_this_execution": (
        cell37_api_calls_this_execution
        == 0
    ),

    "zero_checkpoint_writes_this_execution": (
        cell37_checkpoints_created_this_execution
        == 0
    ),

    "api_execution_disabled": (
        CELL37_ALLOW_API_FOR_MISSING
        is False
    ),

    "full_generation_not_started": True,
}


cell37_runner_status = (
    "PASS"
    if all(
        cell37_runner_checks.values()
    )
    else "FAIL"
)


cell37_semantic_pass_count = (
    cell37_semantic_distribution.get(
        "PASS",
        0,
    )
)

cell37_semantic_fail_count = (
    cell37_semantic_distribution.get(
        "FAIL",
        0,
    )
)


cell37_historical_experimental_status = (
    "PASS"
    if (
        cell37_structural_pass_count
        == 12
        and cell37_semantic_pass_count
        == 12
    )
    else "FAIL"
)


if cell37_runner_status != "PASS":

    cell37_next_readiness = (
        "BLOCKED_CELL37_RUNNER_REPAIR_REQUIRED"
    )

elif (
    cell37_structural_pass_count
    == 12
    and cell37_semantic_pass_count
    == 12
):

    cell37_next_readiness = (
        "HISTORICAL_CANARY_COMPLETE_NO_RECOVERY_REQUIRED"
    )

else:

    cell37_next_readiness = (
        "HISTORICAL_CANARY_COMPLETE_USE_DOWNSTREAM_RECOVERY"
    )


# =========================================================
# OUTPUT
# =========================================================

print()

print(
    "=== GOLDEN V2 QA SEMANTIC CANARY RECONSTRUCTION RESULT ==="
)

print()

print(
    "Results reconstructed:",
    len(cell37_results),
)

print(
    "Valid checkpoints:",
    cell37_valid_checkpoint_count,
)

print(
    "Missing checkpoints:",
    cell37_missing_checkpoint_count,
)

print(
    "Invalid checkpoints:",
    cell37_invalid_checkpoint_count,
)

print(
    "Resolution modes:",
    dict(
        Counter(
            row[
                "resolution_mode"
            ]
            for row in cell37_resolution_modes
        )
    ),
)

print(
    "API calls this execution:",
    cell37_api_calls_this_execution,
)

print(
    "Checkpoints created this execution:",
    cell37_checkpoints_created_this_execution,
)

print()

print(
    "Historical Structural PASS:",
    cell37_structural_pass_count,
    "/ 12",
)

print(
    "Historical Semantic verdict distribution:",
    dict(
        cell37_semantic_distribution
    ),
)

print(
    "Historical dimension failures:",
    dict(
        cell37_dimension_failures
    ),
)

print(
    "Historical estimated cost represented by checkpoints (toman):",
    cell37_total_estimated_cost_toman,
)

print()

print(
    "Runner checks:"
)

for (
    check_name,
    check_value,
) in cell37_runner_checks.items():

    print(
        " -",
        check_name,
        ":",
        (
            "PASS"
            if check_value
            else "FAIL"
        ),
    )

print()

print(
    "Runner health:",
    cell37_runner_status,
)

print(
    "Historical experimental result:",
    cell37_historical_experimental_status,
)

print(
    "Historical Semantic PASS:",
    cell37_semantic_pass_count,
    "/ 12",
)

print(
    "Historical Semantic FAIL:",
    cell37_semantic_fail_count,
    "/ 12",
)

print(
    "API execution allowed:",
    CELL37_ALLOW_API_FOR_MISSING,
)

print(
    "Full Generation started:",
    False,
)

print(
    "Next readiness:",
    cell37_next_readiness,
)

print(
    "CELL 37 CHECKPOINTED RUNNER:",
    cell37_runner_status,
)


# Only reconstruction/checkpoint integrity can fail the cell.
# Historical Canary failures are preserved and handled downstream.
assert cell37_runner_status == "PASS", (
    "Cell 37 historical checkpoint reconstruction failed."
)
````

<h2>مرحله ۳۹ — بازیابی داوری‌های ناقص Semantic Canary</h2>
<p>
تحلیل مرحله قبل نشان داد سه مورد از نتایج اعتبارسنجی معنایی به‌دلیل رسیدن خروجی Validator به محدودیت طول ناقص شده‌اند و شکست معنایی محسوب نمی‌شوند.
در این مرحله فقط همان سه داوری ناقص با سقف خروجی ۱۵۳۶ توکن دوباره اجرا می‌شوند و ۹ داوری ساختاری سالم قبلی بدون اجرای مجدد حفظ می‌شوند.
شکست معنایی واقعیِ شناسایی‌شده نیز بدون تغییر حفظ می‌شود تا پس از تکمیل مجموعه ۱۲ داوری، درباره اصلاح Generation تصمیم‌گیری شود.
Full Generation در این مرحله آغاز نمی‌شود.
</p>
</div>

````python
# Code Cell 39 — Recover Truncated Semantic Canary Validations

from collections import Counter
from pathlib import Path
import hashlib
import json
import os
import re
import tempfile
import time


# =========================================================
# CONFIG
# =========================================================

CELL39_RUNNER_VERSION = (
    "golden_v2_qa_semantic_canary_recovery_v1"
)

CELL39_MAX_TOKENS = 1536
CELL39_TEMPERATURE = 0.0
CELL39_TOP_P = 1.0

CELL39_MAX_ATTEMPTS = 2
CELL39_RETRY_SECONDS = 2.0
CELL39_REQUEST_SPACING_SECONDS = 2.7

# Intentional paid recovery of ONLY truncated validations.
# MUST become False before final Restart -> Run All.
CELL39_ALLOW_API_FOR_MISSING = False


# =========================================================
# REQUIRED STATE
# =========================================================

required_cell39_names = [
    "project_root",
    "semantic_review_client",
    "semantic_review_model",
    "cell37_results",
    "cell37_requests",
    "cell37_skeleton_by_candidate_id",
    "cell37_generation_by_candidate_id",
    "cell37_build_user_prompt",
    "cell37_validate_output",
    "cell37_clean_model_json",
    "cell37_is_rate_limit_exception",
    "CELL37_SYSTEM_PROMPT",
    "CELL37_BOOLEAN_FIELDS",
    "CELL37_RUNNER_VERSION",
    "CELL37_VALIDATION_PROTOCOL_VERSION",
    "CELL37_OUTPUT_SCHEMA_VERSION",
    "GENERATION_V3_VERSION",
    "GENERATION_V3_PROTOCOL_SHA256",
]

missing_cell39_names = [
    name
    for name in required_cell39_names
    if name not in globals()
]

assert not missing_cell39_names, (
    "Missing required Cell 39 state: "
    + ", ".join(missing_cell39_names)
)

assert len(cell37_results) == 12
assert len(cell37_requests) == 12


# =========================================================
# IDENTIFY EXACT TRUNCATION TARGETS
# =========================================================

cell39_recovery_targets = [
    result
    for result in cell37_results
    if (
        result.get("structural_status") != "PASS"
        and result.get("finish_reason") == "length"
        and "JSON_PARSE_FAILED"
        in (
            result.get("structural_errors")
            or []
        )
    )
]

assert len(cell39_recovery_targets) == 3, (
    "Expected exactly 3 truncated semantic validations, got "
    f"{len(cell39_recovery_targets)}."
)

cell39_recovery_candidate_ids = [
    result["candidate_id"]
    for result in cell39_recovery_targets
]

assert len(
    set(cell39_recovery_candidate_ids)
) == 3


cell39_original_request_by_candidate_id = {
    request["candidate_id"]: request
    for request in cell37_requests
}

assert all(
    candidate_id
    in cell39_original_request_by_candidate_id
    for candidate_id
    in cell39_recovery_candidate_ids
)


# =========================================================
# HASH UTILITIES
# =========================================================

def cell39_canonical_json(value):

    return json.dumps(
        value,
        ensure_ascii=False,
        sort_keys=True,
        separators=(",", ":"),
    )


def cell39_sha256_text(text):

    return hashlib.sha256(
        text.encode("utf-8")
    ).hexdigest()


# =========================================================
# REQUEST CONTRACT
# =========================================================

def cell39_build_request(
    candidate_id,
):

    original_request = (
        cell39_original_request_by_candidate_id[
            candidate_id
        ]
    )

    user_prompt = (
        original_request[
            "user_prompt"
        ]
    )

    generated_qa_sha256 = (
        original_request[
            "generated_qa_sha256"
        ]
    )

    fingerprint_payload = {
        "runner_version": (
            CELL39_RUNNER_VERSION
        ),

        "validation_protocol_version": (
            CELL37_VALIDATION_PROTOCOL_VERSION
        ),

        "output_schema_version": (
            CELL37_OUTPUT_SCHEMA_VERSION
        ),

        "candidate_id": (
            candidate_id
        ),

        "generation_protocol_version": (
            GENERATION_V3_VERSION
        ),

        "generation_protocol_sha256": (
            GENERATION_V3_PROTOCOL_SHA256
        ),

        "generated_qa_sha256": (
            generated_qa_sha256
        ),

        "model": (
            semantic_review_model
        ),

        "system_prompt": (
            CELL37_SYSTEM_PROMPT
        ),

        "user_prompt": (
            user_prompt
        ),

        "temperature": (
            CELL39_TEMPERATURE
        ),

        "top_p": (
            CELL39_TOP_P
        ),

        "max_tokens": (
            CELL39_MAX_TOKENS
        ),

        "enable_thinking": False,
    }

    request_fingerprint = (
        cell39_sha256_text(
            cell39_canonical_json(
                fingerprint_payload
            )
        )
    )

    return {
        "candidate_id": (
            candidate_id
        ),

        "user_prompt": (
            user_prompt
        ),

        "generated_qa_sha256": (
            generated_qa_sha256
        ),

        "request_fingerprint": (
            request_fingerprint
        ),
    }


cell39_requests = [
    cell39_build_request(
        candidate_id
    )
    for candidate_id
    in cell39_recovery_candidate_ids
]

assert len(cell39_requests) == 3


# =========================================================
# CHECKPOINT DIRECTORY
# =========================================================

cell39_checkpoint_dir = (
    Path(project_root)
    / "artifacts"
    / "phase2"
    / "golden_dataset_v2"
    / "work"
    / "qa_semantic_canary_recovery_1536_v1"
)

cell39_checkpoint_dir.mkdir(
    parents=True,
    exist_ok=True,
)


def cell39_checkpoint_path(
    candidate_id,
):

    safe_id = re.sub(
        r"[^A-Za-z0-9_.-]+",
        "_",
        candidate_id,
    )

    return (
        cell39_checkpoint_dir
        / f"{safe_id}.json"
    )


# =========================================================
# ATOMIC WRITE
# =========================================================

def cell39_atomic_write_json(
    path,
    payload,
):

    path = Path(path)

    content = (
        json.dumps(
            payload,
            ensure_ascii=False,
            sort_keys=True,
            indent=2,
        )
        + "\n"
    )

    fd, temp_name = tempfile.mkstemp(
        prefix=path.name + ".",
        suffix=".tmp",
        dir=str(path.parent),
    )

    try:

        with os.fdopen(
            fd,
            "w",
            encoding="utf-8",
        ) as handle:

            handle.write(content)
            handle.flush()
            os.fsync(
                handle.fileno()
            )

        os.replace(
            temp_name,
            path,
        )

    except Exception:

        try:
            os.unlink(temp_name)
        except FileNotFoundError:
            pass

        raise


# =========================================================
# CHECKPOINT CONTRACT
# =========================================================

def cell39_build_checkpoint(
    request,
    result,
):

    checkpoint = {
        "checkpoint_version": (
            CELL39_RUNNER_VERSION
        ),

        "candidate_id": (
            request[
                "candidate_id"
            ]
        ),

        "request_fingerprint": (
            request[
                "request_fingerprint"
            ]
        ),

        "generated_qa_sha256": (
            request[
                "generated_qa_sha256"
            ]
        ),

        "validation_protocol_version": (
            CELL37_VALIDATION_PROTOCOL_VERSION
        ),

        "output_schema_version": (
            CELL37_OUTPUT_SCHEMA_VERSION
        ),

        "model": (
            semantic_review_model
        ),

        "max_tokens": (
            CELL39_MAX_TOKENS
        ),

        "result": result,
    }

    checkpoint[
        "checkpoint_payload_sha256"
    ] = cell39_sha256_text(
        cell39_canonical_json(
            checkpoint
        )
    )

    return checkpoint


def cell39_validate_checkpoint(
    checkpoint,
    request,
):

    errors = []

    if not isinstance(
        checkpoint,
        dict,
    ):
        return [
            "CHECKPOINT_NOT_OBJECT"
        ]

    stored_sha = checkpoint.get(
        "checkpoint_payload_sha256"
    )

    payload_without_sha = {
        key: value
        for key, value
        in checkpoint.items()
        if key
        != "checkpoint_payload_sha256"
    }

    recalculated_sha = (
        cell39_sha256_text(
            cell39_canonical_json(
                payload_without_sha
            )
        )
    )

    if stored_sha != recalculated_sha:
        errors.append(
            "CHECKPOINT_HASH_MISMATCH"
        )

    exact_expectations = {
        "checkpoint_version": (
            CELL39_RUNNER_VERSION
        ),

        "candidate_id": (
            request[
                "candidate_id"
            ]
        ),

        "request_fingerprint": (
            request[
                "request_fingerprint"
            ]
        ),

        "generated_qa_sha256": (
            request[
                "generated_qa_sha256"
            ]
        ),

        "validation_protocol_version": (
            CELL37_VALIDATION_PROTOCOL_VERSION
        ),

        "output_schema_version": (
            CELL37_OUTPUT_SCHEMA_VERSION
        ),

        "model": (
            semantic_review_model
        ),

        "max_tokens": (
            CELL39_MAX_TOKENS
        ),
    }

    for (
        field_name,
        expected_value,
    ) in exact_expectations.items():

        if (
            checkpoint.get(
                field_name
            )
            != expected_value
        ):
            errors.append(
                f"{field_name.upper()}_MISMATCH"
            )

    result = checkpoint.get(
        "result"
    )

    if not isinstance(
        result,
        dict,
    ):

        errors.append(
            "CHECKPOINT_RESULT_INVALID"
        )

    elif (
        result.get(
            "candidate_id"
        )
        != request[
            "candidate_id"
        ]
    ):

        errors.append(
            "CHECKPOINT_RESULT_CANDIDATE_MISMATCH"
        )

    return errors


def cell39_load_checkpoint(
    request,
):

    path = cell39_checkpoint_path(
        request[
            "candidate_id"
        ]
    )

    if not path.exists():

        return (
            None,
            "MISSING",
            [],
        )

    try:

        with path.open(
            "r",
            encoding="utf-8",
        ) as handle:

            checkpoint = json.load(
                handle
            )

    except Exception as exc:

        return (
            None,
            "UNREADABLE",
            [
                f"{type(exc).__name__}: {exc}"
            ],
        )

    errors = (
        cell39_validate_checkpoint(
            checkpoint,
            request,
        )
    )

    if errors:

        return (
            checkpoint,
            "INVALID",
            errors,
        )

    return (
        checkpoint,
        "VALID",
        [],
    )


# =========================================================
# COST
# =========================================================

cell39_has_cost_rates = (
    "INPUT_TOMAN_PER_TOKEN" in globals()
    and
    "OUTPUT_TOMAN_PER_TOKEN" in globals()
)


def cell39_estimated_cost_toman(
    input_tokens,
    output_tokens,
):

    if not cell39_has_cost_rates:
        return None

    return (
        input_tokens
        * INPUT_TOMAN_PER_TOKEN
        +
        output_tokens
        * OUTPUT_TOMAN_PER_TOKEN
    )


# =========================================================
# API EXECUTION
# =========================================================

def cell39_execute_request(
    request,
):

    attempt_history = []

    for attempt_number in range(
        1,
        CELL39_MAX_ATTEMPTS + 1,
    ):

        started_at = time.time()

        try:

            response = (
                semantic_review_client
                .chat
                .completions
                .create(
                    model=(
                        semantic_review_model
                    ),

                    messages=[
                        {
                            "role": "system",
                            "content": (
                                CELL37_SYSTEM_PROMPT
                            ),
                        },
                        {
                            "role": "user",
                            "content": (
                                request[
                                    "user_prompt"
                                ]
                            ),
                        },
                    ],

                    temperature=(
                        CELL39_TEMPERATURE
                    ),

                    top_p=(
                        CELL39_TOP_P
                    ),

                    max_tokens=(
                        CELL39_MAX_TOKENS
                    ),

                    extra_body={
                        "enable_thinking": False,
                    },
                )
            )

        except Exception as exc:

            attempt_history.append({
                "attempt": (
                    attempt_number
                ),

                "status": (
                    "API_ERROR"
                ),

                "error": (
                    f"{type(exc).__name__}: {exc}"
                ),
            })

            if (
                attempt_number
                < CELL39_MAX_ATTEMPTS
            ):

                wait_seconds = (
                    70.0
                    if cell37_is_rate_limit_exception(
                        exc
                    )
                    else (
                        CELL39_RETRY_SECONDS
                        * attempt_number
                    )
                )

                time.sleep(
                    wait_seconds
                )

                continue

            return {
                "candidate_id": (
                    request[
                        "candidate_id"
                    ]
                ),

                "execution_status": (
                    "API_FAILED"
                ),

                "attempt_history": (
                    attempt_history
                ),
            }

        elapsed_seconds = (
            time.time()
            - started_at
        )

        choice = response.choices[0]
        message = choice.message

        finish_reason = getattr(
            choice,
            "finish_reason",
            None,
        )

        raw_content = getattr(
            message,
            "content",
            None,
        )

        usage = getattr(
            response,
            "usage",
            None,
        )

        input_tokens = int(
            getattr(
                usage,
                "prompt_tokens",
                0,
            )
            or 0
        )

        output_tokens = int(
            getattr(
                usage,
                "completion_tokens",
                0,
            )
            or 0
        )

        total_tokens = int(
            getattr(
                usage,
                "total_tokens",
                0,
            )
            or (
                input_tokens
                + output_tokens
            )
        )

        reasoning_tokens = 0

        completion_details = getattr(
            usage,
            "completion_tokens_details",
            None,
        )

        if completion_details is not None:

            reasoning_tokens = int(
                getattr(
                    completion_details,
                    "reasoning_tokens",
                    0,
                )
                or 0
            )

        clean_content = (
            cell37_clean_model_json(
                raw_content
            )
        )

        parsed_json = None
        parse_error = None

        try:

            parsed_json = json.loads(
                clean_content
            )

        except Exception as exc:

            parse_error = (
                f"{type(exc).__name__}: {exc}"
            )

        if parsed_json is None:

            structural_errors = [
                "JSON_PARSE_FAILED"
            ]

        else:

            structural_errors = (
                cell37_validate_output(
                    parsed_json
                )
            )

        return {
            "candidate_id": (
                request[
                    "candidate_id"
                ]
            ),

            "request_fingerprint": (
                request[
                    "request_fingerprint"
                ]
            ),

            "generated_qa_sha256": (
                request[
                    "generated_qa_sha256"
                ]
            ),

            "model": (
                semantic_review_model
            ),

            "finish_reason": (
                finish_reason
            ),

            "elapsed_seconds": round(
                elapsed_seconds,
                4,
            ),

            "usage": {
                "input_tokens": (
                    input_tokens
                ),

                "output_tokens": (
                    output_tokens
                ),

                "total_tokens": (
                    total_tokens
                ),

                "reasoning_tokens": (
                    reasoning_tokens
                ),
            },

            "estimated_cost_toman": (
                cell39_estimated_cost_toman(
                    input_tokens,
                    output_tokens,
                )
            ),

            "raw_content": (
                raw_content
            ),

            "parsed_json": (
                parsed_json
            ),

            "parse_error": (
                parse_error
            ),

            "structural_errors": (
                structural_errors
            ),

            "structural_status": (
                "PASS"
                if not structural_errors
                else "FAIL"
            ),

            "semantic_verdict": (
                parsed_json.get(
                    "verdict"
                )
                if (
                    isinstance(
                        parsed_json,
                        dict,
                    )
                    and not structural_errors
                )
                else None
            ),

            "attempt_history": (
                attempt_history
            ),
        }

    raise RuntimeError(
        "Unexpected Cell 39 execution path."
    )


# =========================================================
# PREFLIGHT
# =========================================================

print(
    "=== GOLDEN V2 SEMANTIC CANARY RECOVERY PREFLIGHT ==="
)

print()

print(
    "Recovery population:",
    len(cell39_requests),
)

print(
    "Recovery candidates:",
    cell39_recovery_candidate_ids,
)

print(
    "Model:",
    semantic_review_model,
)

print(
    "Max tokens:",
    CELL39_MAX_TOKENS,
)

print(
    "API allowed for missing exact checkpoints:",
    CELL39_ALLOW_API_FOR_MISSING,
)

print(
    "Full Generation started:",
    False,
)

print()


# =========================================================
# RESOLVE RECOVERY TARGETS
# =========================================================

cell39_recovery_results = []
cell39_resolution_modes = []

cell39_api_calls_this_execution = 0
cell39_checkpoints_created_this_execution = 0


for index, request in enumerate(
    cell39_requests,
    start=1,
):

    (
        checkpoint,
        checkpoint_status,
        checkpoint_errors,
    ) = cell39_load_checkpoint(
        request
    )

    candidate_id = request[
        "candidate_id"
    ]

    if checkpoint_status == "VALID":

        result = checkpoint[
            "result"
        ]

        resolution_mode = (
            "EXACT_CHECKPOINT"
        )

    else:

        if checkpoint_status in {
            "INVALID",
            "UNREADABLE",
        }:

            raise AssertionError(
                "Existing Cell 39 checkpoint is invalid. "
                "Refusing automatic paid replacement for "
                f"{candidate_id}. "
                f"Status={checkpoint_status}; "
                f"errors={checkpoint_errors}"
            )

        if not CELL39_ALLOW_API_FOR_MISSING:

            raise AssertionError(
                "Exact Cell 39 checkpoint is missing and "
                "API execution is disabled for "
                f"{candidate_id}."
            )

        if (
            cell39_api_calls_this_execution
            > 0
        ):

            time.sleep(
                CELL39_REQUEST_SPACING_SECONDS
            )

        result = (
            cell39_execute_request(
                request
            )
        )

        cell39_api_calls_this_execution += 1

        checkpoint = (
            cell39_build_checkpoint(
                request,
                result,
            )
        )

        cell39_atomic_write_json(
            cell39_checkpoint_path(
                candidate_id
            ),
            checkpoint,
        )

        (
            verified_checkpoint,
            verified_status,
            verified_errors,
        ) = cell39_load_checkpoint(
            request
        )

        assert (
            verified_status
            == "VALID"
        ), (
            "New Cell 39 checkpoint verification failed "
            f"for {candidate_id}: "
            f"{verified_errors}"
        )

        result = (
            verified_checkpoint[
                "result"
            ]
        )

        resolution_mode = (
            "NEW_PAID_EXECUTION"
        )

        cell39_checkpoints_created_this_execution += 1

    cell39_recovery_results.append(
        result
    )

    cell39_resolution_modes.append({
        "candidate_id": (
            candidate_id
        ),

        "resolution_mode": (
            resolution_mode
        ),
    })

    print(
        f"[{index:02d}/03]",
        candidate_id,
        "|",
        resolution_mode,
        "| structural=",
        result.get(
            "structural_status",
            result.get(
                "execution_status"
            ),
        ),
        "| semantic=",
        result.get(
            "semantic_verdict"
        ),
        flush=True,
    )


# =========================================================
# FINAL DISK RESCAN
# =========================================================

cell39_valid_checkpoint_count = 0
cell39_missing_checkpoint_count = 0
cell39_invalid_checkpoint_count = 0


for request in cell39_requests:

    (
        checkpoint,
        status,
        errors,
    ) = cell39_load_checkpoint(
        request
    )

    if status == "VALID":

        cell39_valid_checkpoint_count += 1

    elif status == "MISSING":

        cell39_missing_checkpoint_count += 1

    else:

        cell39_invalid_checkpoint_count += 1


# =========================================================
# RECONSTRUCT FINAL 12 VALIDATIONS
# =========================================================

cell39_recovery_by_candidate_id = {
    result["candidate_id"]: result
    for result in cell39_recovery_results
}

assert len(
    cell39_recovery_by_candidate_id
) == 3


cell39_final_validation_results = []

for original_result in cell37_results:

    candidate_id = (
        original_result[
            "candidate_id"
        ]
    )

    if (
        candidate_id
        in cell39_recovery_by_candidate_id
    ):

        cell39_final_validation_results.append(
            cell39_recovery_by_candidate_id[
                candidate_id
            ]
        )

    else:

        cell39_final_validation_results.append(
            original_result
        )


assert len(
    cell39_final_validation_results
) == 12

assert len({
    result["candidate_id"]
    for result
    in cell39_final_validation_results
}) == 12


# =========================================================
# FINAL COUNTS
# =========================================================

cell39_final_structural_pass_count = sum(
    result.get(
        "structural_status"
    )
    == "PASS"
    for result
    in cell39_final_validation_results
)

cell39_final_semantic_distribution = Counter(
    result.get(
        "semantic_verdict"
    )
    for result
    in cell39_final_validation_results
    if result.get(
        "structural_status"
    )
    == "PASS"
)

cell39_final_dimension_failures = Counter()

for result in cell39_final_validation_results:

    if result.get(
        "structural_status"
    ) != "PASS":
        continue

    parsed = result.get(
        "parsed_json"
    )

    if not isinstance(
        parsed,
        dict,
    ):
        continue

    for field_name in CELL37_BOOLEAN_FIELDS:

        if parsed.get(
            field_name
        ) is False:

            cell39_final_dimension_failures[
                field_name
            ] += 1


cell39_api_failure_count = sum(
    result.get(
        "execution_status"
    )
    == "API_FAILED"
    for result
    in cell39_recovery_results
)

cell39_known_costs = [
    result.get(
        "estimated_cost_toman"
    )
    for result
    in cell39_recovery_results
    if isinstance(
        result.get(
            "estimated_cost_toman"
        ),
        (int, float),
    )
]

cell39_total_estimated_cost_toman = (
    sum(
        cell39_known_costs
    )
    if cell39_known_costs
    else None
)


# =========================================================
# RUNNER HEALTH
# =========================================================

cell39_runner_checks = {
    "3_recovery_targets": (
        len(cell39_requests)
        == 3
    ),

    "3_results_reconstructed": (
        len(cell39_recovery_results)
        == 3
    ),

    "3_valid_recovery_checkpoints": (
        cell39_valid_checkpoint_count
        == 3
    ),

    "zero_missing_recovery_checkpoints": (
        cell39_missing_checkpoint_count
        == 0
    ),

    "zero_invalid_recovery_checkpoints": (
        cell39_invalid_checkpoint_count
        == 0
    ),

    "zero_api_failures": (
        cell39_api_failure_count
        == 0
    ),

    "final_population_12": (
        len(
            cell39_final_validation_results
        )
        == 12
    ),

    "full_generation_not_started": True,
}

cell39_runner_status = (
    "PASS"
    if all(
        cell39_runner_checks.values()
    )
    else "FAIL"
)


cell39_semantic_pass_count = (
    cell39_final_semantic_distribution.get(
        "PASS",
        0,
    )
)

cell39_semantic_fail_count = (
    cell39_final_semantic_distribution.get(
        "FAIL",
        0,
    )
)


if (
    cell39_runner_status == "PASS"
    and cell39_final_structural_pass_count
    == 12
):

    cell39_next_readiness = (
        "READY_FOR_FINAL_QA_CANARY_SEMANTIC_ADJUDICATION"
    )

else:

    cell39_next_readiness = (
        "READY_FOR_CELL39_FAILURE_ANALYSIS"
        if cell39_runner_status == "PASS"
        else "BLOCKED_CELL39_RUNNER_REPAIR_REQUIRED"
    )


# =========================================================
# OUTPUT
# =========================================================

print()

print(
    "=== GOLDEN V2 SEMANTIC CANARY RECOVERY RESULT ==="
)

print()

print(
    "Recovery results:",
    len(cell39_recovery_results),
)

print(
    "Valid recovery checkpoints:",
    cell39_valid_checkpoint_count,
)

print(
    "Missing recovery checkpoints:",
    cell39_missing_checkpoint_count,
)

print(
    "Invalid recovery checkpoints:",
    cell39_invalid_checkpoint_count,
)

print(
    "Resolution modes:",
    dict(
        Counter(
            row[
                "resolution_mode"
            ]
            for row
            in cell39_resolution_modes
        )
    ),
)

print(
    "API calls this execution:",
    cell39_api_calls_this_execution,
)

print(
    "Checkpoints created this execution:",
    cell39_checkpoints_created_this_execution,
)

print(
    "Estimated recovery cost (toman):",
    cell39_total_estimated_cost_toman,
)

print()

print(
    "Final Structural PASS:",
    cell39_final_structural_pass_count,
    "/ 12",
)

print(
    "Final Semantic distribution:",
    dict(
        cell39_final_semantic_distribution
    ),
)

print(
    "Final dimension failures:",
    dict(
        cell39_final_dimension_failures
    ),
)

print()

print(
    "Runner checks:"
)

for (
    check_name,
    check_value,
) in cell39_runner_checks.items():

    print(
        " -",
        check_name,
        ":",
        (
            "PASS"
            if check_value
            else "FAIL"
        ),
    )

print()

print(
    "Runner health:",
    cell39_runner_status,
)

print(
    "Semantic PASS:",
    cell39_semantic_pass_count,
    "/ 12",
)

print(
    "Semantic FAIL:",
    cell39_semantic_fail_count,
    "/ 12",
)

print(
    "Full Generation started:",
    False,
)

print(
    "Next readiness:",
    cell39_next_readiness,
)

print(
    "CELL 39 CHECKPOINTED RUNNER:",
    cell39_runner_status,
)

assert cell39_runner_status == "PASS", (
    "Cell 39 checkpointed runner health failed."
)
````

<div style="direction: rtl; white-space: normal; line-height: 1.7;">
<h2>مرحله ۴۰ — داوری نهایی شکست‌های معنایی Canary</h2>
<p>
پس از بازیابی کامل خروجی Validator، هر ۱۲ داوری از نظر ساختاری معتبر هستند و دو مورد دارای شکست معنایی واقعی‌اند.
در این مرحله فقط همین دو مورد با نمایش Claim، Evidence، Question، Expected Answer و تمام ابعاد داوری بررسی می‌شوند تا مشخص شود ایراد دقیق Generation چیست و آیا اصلاح قرارداد تولید پیش از Full Generation لازم است.
این مرحله هیچ فراخوانی API، تولید مجدد یا تغییر فایل انجام نمی‌دهد.
</p>
</div>

````python
# Code Cell 40 — Final QA Canary Semantic Failure Adjudication

from collections import Counter
import json


# =========================================================
# REQUIRED STATE
# =========================================================

required_cell40_names = [
    "cell39_final_validation_results",
    "cell37_skeleton_by_candidate_id",
    "cell37_generation_by_candidate_id",
    "CELL37_BOOLEAN_FIELDS",
]

missing_cell40_names = [
    name
    for name in required_cell40_names
    if name not in globals()
]

assert not missing_cell40_names, (
    "Missing required Cell 40 state: "
    + ", ".join(missing_cell40_names)
)

assert len(cell39_final_validation_results) == 12


# =========================================================
# VERIFY FINAL CANARY STATE
# =========================================================

cell40_structural_failures = [
    result
    for result in cell39_final_validation_results
    if result.get("structural_status") != "PASS"
]

assert len(cell40_structural_failures) == 0, (
    "Cell 40 requires 12/12 structurally valid semantic reviews."
)


cell40_semantic_passes = [
    result
    for result in cell39_final_validation_results
    if result.get("semantic_verdict") == "PASS"
]

cell40_semantic_failures = [
    result
    for result in cell39_final_validation_results
    if result.get("semantic_verdict") == "FAIL"
]


assert len(cell40_semantic_passes) == 10, (
    "Expected exactly 10 semantic PASS results, got "
    f"{len(cell40_semantic_passes)}."
)

assert len(cell40_semantic_failures) == 2, (
    "Expected exactly 2 semantic FAIL results, got "
    f"{len(cell40_semantic_failures)}."
)


# =========================================================
# BUILD FAILURE AUDIT
# =========================================================

cell40_failure_audit = []

for validation_result in cell40_semantic_failures:

    candidate_id = validation_result["candidate_id"]

    assert (
        candidate_id
        in cell37_skeleton_by_candidate_id
    )

    assert (
        candidate_id
        in cell37_generation_by_candidate_id
    )

    skeleton = (
        cell37_skeleton_by_candidate_id[
            candidate_id
        ]
    )

    generation_result = (
        cell37_generation_by_candidate_id[
            candidate_id
        ]
    )

    generated_qa = generation_result.get(
        "parsed_json"
    )

    validation = validation_result.get(
        "parsed_json"
    )

    assert isinstance(
        generated_qa,
        dict,
    )

    assert isinstance(
        validation,
        dict,
    )

    evidence = []

    for (
        evidence_id,
        evidence_quote,
        evidence_locator,
    ) in zip(
        skeleton["evidence_ids"],
        skeleton["evidence_quotes"],
        skeleton["evidence_locators"],
    ):

        evidence.append({
            "evidence_id": evidence_id,
            "quote": evidence_quote,
            "locator": evidence_locator,
        })

    failed_dimensions = [
        field_name
        for field_name
        in CELL37_BOOLEAN_FIELDS
        if validation.get(
            field_name
        ) is False
    ]

    cell40_failure_audit.append({
        "candidate_id": candidate_id,

        "claim_id": (
            skeleton["claim_id"]
        ),

        "claim_text_fa": (
            skeleton["claim_text_fa"]
        ),

        "evidence": evidence,

        "question": (
            generated_qa["question"]
        ),

        "expected_answer": (
            generated_qa[
                "expected_answer"
            ]
        ),

        "question_intent": (
            generated_qa[
                "question_intent"
            ]
        ),

        "semantic_verdict": (
            validation["verdict"]
        ),

        "failed_dimensions": (
            failed_dimensions
        ),

        "failure_reasons": (
            validation[
                "failure_reasons"
            ]
        ),

        "full_validation": (
            validation
        ),
    })


assert len(cell40_failure_audit) == 2


# =========================================================
# DETERMINISTIC FAILURE CLASSIFICATION
# =========================================================

cell40_failure_classes = Counter()

for audit in cell40_failure_audit:

    failed = set(
        audit["failed_dimensions"]
    )

    classes = []

    if (
        "question_targets_claim" in failed
        or
        "answer_question_alignment" in failed
    ):
        classes.append(
            "QUESTION_TARGET_OR_ALIGNMENT_DRIFT"
        )

    if (
        "scope_fidelity" in failed
        or
        "modality_quantifier_fidelity" in failed
    ):
        classes.append(
            "CLAIM_FIDELITY_DRIFT"
        )

    if (
        "no_neighbor_fact_drift" in failed
    ):
        classes.append(
            "NEIGHBOR_FACT_DRIFT"
        )

    if (
        "no_external_knowledge" in failed
    ):
        classes.append(
            "EXTERNAL_KNOWLEDGE_ADDITION"
        )

    if (
        "standalone_question" in failed
    ):
        classes.append(
            "NON_STANDALONE_QUESTION"
        )

    if (
        "answer_recovers_claim" in failed
    ):
        classes.append(
            "ANSWER_DOES_NOT_RECOVER_CLAIM"
        )

    if not classes:
        classes.append(
            "OTHER_SEMANTIC_FAILURE"
        )

    audit[
        "failure_classes"
    ] = classes

    for failure_class in classes:

        cell40_failure_classes[
            failure_class
        ] += 1


# =========================================================
# DETERMINE PROTOCOL IMPLICATION
# =========================================================

cell40_question_target_failure_count = sum(
    "question_targets_claim"
    in audit["failed_dimensions"]
    for audit in cell40_failure_audit
)

cell40_answer_alignment_failure_count = sum(
    "answer_question_alignment"
    in audit["failed_dimensions"]
    for audit in cell40_failure_audit
)

cell40_fidelity_failure_count = sum(
    (
        "scope_fidelity"
        in audit["failed_dimensions"]
    )
    or
    (
        "modality_quantifier_fidelity"
        in audit["failed_dimensions"]
    )
    for audit in cell40_failure_audit
)


# Any real semantic failure in the Canary blocks scaling.
cell40_generation_protocol_revision_required = (
    len(cell40_semantic_failures) > 0
)

cell40_full_generation_allowed = False


# =========================================================
# PRINT SUMMARY
# =========================================================

print(
    "=== GOLDEN V2 FINAL QA CANARY SEMANTIC ADJUDICATION ==="
)

print()

print(
    "Final Canary population:",
    len(cell39_final_validation_results),
)

print(
    "Structural PASS:",
    12 - len(cell40_structural_failures),
    "/ 12",
)

print(
    "Semantic PASS:",
    len(cell40_semantic_passes),
    "/ 12",
)

print(
    "Semantic FAIL:",
    len(cell40_semantic_failures),
    "/ 12",
)

print()

print(
    "Failure class distribution:",
    dict(cell40_failure_classes),
)

print(
    "Question target failures:",
    cell40_question_target_failure_count,
)

print(
    "Answer alignment failures:",
    cell40_answer_alignment_failure_count,
)

print(
    "Scope/modality/quantifier fidelity failures:",
    cell40_fidelity_failure_count,
)


# =========================================================
# PRINT BOTH FAILURES IN FULL
# =========================================================

for index, audit in enumerate(
    cell40_failure_audit,
    start=1,
):

    print()
    print("=" * 78)

    print(
        f"SEMANTIC FAILURE {index} / "
        f"{len(cell40_failure_audit)}"
    )

    print("=" * 78)

    print(
        "Candidate ID:",
        audit["candidate_id"],
    )

    print(
        "Claim ID:",
        audit["claim_id"],
    )

    print()

    print("CLAIM:")
    print(
        audit["claim_text_fa"]
    )

    print()

    print("EVIDENCE:")

    for evidence_index, evidence in enumerate(
        audit["evidence"],
        start=1,
    ):

        print(
            f"[{evidence_index}]",
            evidence["quote"],
        )

    print()

    print("QUESTION:")
    print(
        audit["question"]
    )

    print()

    print("EXPECTED ANSWER:")
    print(
        audit["expected_answer"]
    )

    print()

    print("QUESTION INTENT:")
    print(
        audit["question_intent"]
    )

    print()

    print(
        "FAILED DIMENSIONS:",
        audit["failed_dimensions"],
    )

    print(
        "FAILURE CLASSES:",
        audit["failure_classes"],
    )

    print(
        "FAILURE REASONS:",
        audit["failure_reasons"],
    )

    print()

    print("FULL VALIDATION:")

    print(
        json.dumps(
            audit["full_validation"],
            ensure_ascii=False,
            indent=2,
        )
    )


# =========================================================
# FINAL CHECKS
# =========================================================

cell40_checks = {
    "12_final_reviews_available": (
        len(
            cell39_final_validation_results
        )
        == 12
    ),

    "12_structural_pass": (
        len(cell40_structural_failures)
        == 0
    ),

    "10_semantic_pass_confirmed": (
        len(cell40_semantic_passes)
        == 10
    ),

    "2_semantic_fail_confirmed": (
        len(cell40_semantic_failures)
        == 2
    ),

    "2_failures_fully_audited": (
        len(cell40_failure_audit)
        == 2
    ),

    "full_generation_not_started": (
        cell40_full_generation_allowed
        is False
    ),

    "zero_api_calls": True,

    "zero_file_writes": True,
}


print()
print("=" * 78)
print("Validation checks:")

for (
    check_name,
    check_value,
) in cell40_checks.items():

    print(
        " -",
        check_name,
        ":",
        (
            "PASS"
            if check_value
            else "FAIL"
        ),
    )


cell40_status = (
    "PASS"
    if all(
        cell40_checks.values()
    )
    else "FAIL"
)


print()

print(
    "Generation protocol revision required:",
    cell40_generation_protocol_revision_required,
)

print(
    "Full Generation allowed:",
    cell40_full_generation_allowed,
)

print(
    "API calls performed:",
    0,
)

print(
    "Files modified:",
    0,
)

print(
    "CELL 40 STATUS:",
    cell40_status,
)

print(
    "Next readiness:",
    (
        "READY_FOR_GENERATION_V3_PROTOCOL_REPAIR"
        if (
            cell40_status == "PASS"
            and cell40_generation_protocol_revision_required
        )
        else (
            "READY_FOR_FULL_QA_GENERATION_V3"
            if cell40_status == "PASS"
            else "BLOCKED_CELL40_REPAIR_REQUIRED"
        )
    ),
)

assert cell40_status == "PASS"
````

در این مرحله قرارداد Generation به نسخۀ v4 ارتقا می‌یابد تا نوع پرسش مستقیماً از محتوای Claim تبعیت کند و تمام قیود معنایی مؤثر، از جمله کمیت‌ها، درصدها، مقایسه‌ها، نفی، انحصار، شرط، احتمال و قطعیت حفظ شوند.
این مرحله فقط قرارداد و درخواست‌های اصلاحی را می‌سازد و هیچ فراخوانی API انجام نمی‌دهد.

````python
# Code Cell 41 — QA Generation v4 Protocol Repair

from pathlib import Path
import hashlib
import json


# =========================================================
# REQUIRED STATE
# =========================================================

required_cell41_names = [
    "project_root",
    "semantic_review_model",
    "cell25_candidate_skeletons",
    "cell40_failure_audit",
    "GENERATION_V3_VERSION",
    "GENERATION_V3_PROTOCOL_SHA256",
]

missing_cell41_names = [
    name
    for name in required_cell41_names
    if name not in globals()
]

assert not missing_cell41_names, (
    "Missing required Cell 41 state: "
    + ", ".join(missing_cell41_names)
)

assert len(cell25_candidate_skeletons) == 3813
assert len(cell40_failure_audit) == 2


# =========================================================
# VERSION
# =========================================================

GENERATION_V4_VERSION = (
    "golden_v2_question_answer_generation_v4"
)

GENERATION_V4_PROMPT_VERSION = (
    "golden_v2_question_answer_generation_prompt_v4"
)

GENERATION_V4_OUTPUT_SCHEMA_VERSION = (
    "golden_v2_question_answer_generation_output_v1"
)


# =========================================================
# OUTPUT CONTRACT
# =========================================================

GENERATION_V4_OUTPUT_FIELDS = {
    "question",
    "expected_answer",
    "question_intent",
}


# =========================================================
# SYSTEM PROMPT
# =========================================================

GOLDEN_V2_GENERATION_SYSTEM_PROMPT_V4 = """
You generate exactly one Persian Question and one Persian Expected Answer
for a pre-validated Golden Dataset candidate.

ARCHITECTURE

The target Claim and its Evidence have already been validated upstream.
Do not re-evaluate whether the Claim is true.
Do not output a support verdict.
Do not output an evidence verdict.
Do not repair or expand the Claim.
Do not use external knowledge.

Your task is only to transform the target Claim into one faithful,
standalone Question and Expected Answer.

PRIMARY RULE

The Question must ask for exactly the information asserted by the target
Claim, and the Expected Answer must recover the meaning of that Claim.

QUESTION-TYPE FIDELITY

Choose the question type from what the Claim itself asserts.

Do NOT ask:
- why / چرا
- how / چگونه / چطور
- for a cause
- for a reason
- for a mechanism
- for a consequence
- for a recommendation

unless that relation is explicitly asserted by the target Claim itself.

For example, if the Claim only states that something was selected,
exists, equals a value, has a property, or occurred, ask about that fact.
Do not transform it into a question asking why it happened.

The Question must not demand information that the Expected Answer cannot
supply directly from the target Claim.

CLAIM FIDELITY

Preserve every meaning-bearing restriction in the Claim.

This includes, whenever present:

- negation
- only / تنها / فقط
- all / همه
- some / برخی
- many / بسیاری
- most / اکثر
- at least / حداقل
- at most / حداکثر
- exactly / دقیقاً
- percentages
- quantities
- numerical values
- comparisons
- greater-than or less-than relations
- conditions
- exceptions
- temporal restrictions
- population restrictions
- domain restrictions
- possibility
- probability
- necessity
- certainty
- uncertainty
- modality
- perspective or attribution

Do not silently remove, weaken, strengthen, generalize, or invent any
meaning-bearing qualifier.

If the Claim says "تنها 52 درصد", the generated Question/Expected Answer
must preserve the meaning of "تنها"; returning only "52 درصد" is not
sufficient.

ANSWER FIDELITY

The Expected Answer must answer the generated Question directly.

The Expected Answer must preserve the target Claim's substantive meaning
and its meaning-bearing qualifiers.

It may be phrased naturally, but it must not add a neighboring fact merely
because that fact appears in the Evidence.

EVIDENCE USE

Evidence is provided only to ground wording and prevent drift.

Do not replace the target Claim with another statement from the Evidence.
Do not add external facts.
Do not add explanations, causes, examples, names, institutions, studies,
or context unless they are required by the target Claim itself.

STANDALONE REQUIREMENT

The Question must be understandable by itself.

Do not refer to:
- "the claim"
- "the evidence"
- "the text above"
- "the passage"
- hidden dataset context

OUTPUT

Return JSON only.

Return exactly these three fields:

{
  "question": "...",
  "expected_answer": "...",
  "question_intent": "..."
}

Do not return markdown.
Do not return commentary.
Do not return additional fields.
""".strip()


# =========================================================
# CANONICAL JSON / HASH
# =========================================================

def cell41_canonical_json(value):

    return json.dumps(
        value,
        ensure_ascii=False,
        sort_keys=True,
        separators=(",", ":"),
    )


def cell41_sha256_text(text):

    return hashlib.sha256(
        text.encode("utf-8")
    ).hexdigest()


# =========================================================
# PROTOCOL CONTRACT
# =========================================================

cell41_protocol_contract_v4 = {
    "protocol_version": (
        GENERATION_V4_VERSION
    ),

    "prompt_version": (
        GENERATION_V4_PROMPT_VERSION
    ),

    "output_schema_version": (
        GENERATION_V4_OUTPUT_SCHEMA_VERSION
    ),

    "model": (
        semantic_review_model
    ),

    "target_population": 3813,

    "upstream_generation_protocol": {
        "version": (
            GENERATION_V3_VERSION
        ),

        "sha256": (
            GENERATION_V3_PROTOCOL_SHA256
        ),
    },

    "repair_basis": {
        "semantic_canary_population": 12,
        "semantic_pass": 10,
        "semantic_fail": 2,

        "observed_failure_classes": [
            "QUESTION_TARGET_OR_ALIGNMENT_DRIFT",
            "CLAIM_FIDELITY_DRIFT",
        ],
    },

    "generation_scope": (
        "QUESTION_EXPECTED_ANSWER_ONLY"
    ),

    "claim_evidence_revalidation": False,

    "support_verdict_generation": False,

    "material_assertion_generation": False,

    "question_type_fidelity_required": True,

    "causal_question_requires_causal_claim": True,

    "meaning_bearing_qualifier_preservation_required": True,

    "output_fields": [
        "question",
        "expected_answer",
        "question_intent",
    ],

    "system_prompt": (
        GOLDEN_V2_GENERATION_SYSTEM_PROMPT_V4
    ),
}


GENERATION_V4_PROTOCOL_SHA256 = (
    cell41_sha256_text(
        cell41_canonical_json(
            cell41_protocol_contract_v4
        )
    )
)


# =========================================================
# SKELETON LOOKUP
# =========================================================

cell41_skeleton_by_candidate_id = {
    record["candidate_id"]: record
    for record in cell25_candidate_skeletons
}

assert len(
    cell41_skeleton_by_candidate_id
) == 3813


# =========================================================
# USER PROMPT BUILDER
# =========================================================

def cell41_build_generation_user_prompt_v4(
    candidate_id,
):

    assert (
        candidate_id
        in cell41_skeleton_by_candidate_id
    ), (
        "Unknown candidate_id: "
        f"{candidate_id}"
    )

    skeleton = (
        cell41_skeleton_by_candidate_id[
            candidate_id
        ]
    )

    evidence = []

    for (
        evidence_id,
        evidence_quote,
        evidence_locator,
    ) in zip(
        skeleton["evidence_ids"],
        skeleton["evidence_quotes"],
        skeleton["evidence_locators"],
    ):

        evidence.append({
            "evidence_id": evidence_id,
            "quote": evidence_quote,
            "locator": evidence_locator,
        })

    payload = {
        "candidate_id": candidate_id,

        "target_claim": {
            "claim_id": (
                skeleton["claim_id"]
            ),

            "text_fa": (
                skeleton["claim_text_fa"]
            ),

            "claim_type": (
                skeleton["claim_type"]
            ),

            "claim_types": (
                skeleton["claim_types"]
            ),

            "authority_class": (
                skeleton["authority_class"]
            ),
        },

        "evidence": evidence,

        "generation_requirements": {
            "question_directly_targets_claim": True,

            "question_type_must_be_supported_by_claim": True,

            "causal_question_requires_causal_claim": True,

            "preserve_all_meaning_bearing_qualifiers": True,

            "preserve_negation": True,

            "preserve_quantifiers": True,

            "preserve_modality": True,

            "preserve_percentages_and_numbers": True,

            "preserve_comparisons": True,

            "preserve_conditions": True,

            "preserve_scope": True,

            "no_neighbor_fact_drift": True,

            "no_external_knowledge": True,

            "standalone_question": True,

            "answer_must_directly_answer_question": True,
        },
    }

    return (
        "Generate exactly one faithful Persian Question and "
        "Expected Answer for this pre-validated target.\n\n"
        + json.dumps(
            payload,
            ensure_ascii=False,
            sort_keys=True,
            indent=2,
        )
    )


# =========================================================
# OUTPUT VALIDATOR
# =========================================================

def cell41_validate_generation_output_v4(
    parsed_json,
):

    errors = []

    if not isinstance(
        parsed_json,
        dict,
    ):
        return [
            "OUTPUT_NOT_OBJECT"
        ]

    if (
        set(parsed_json.keys())
        != GENERATION_V4_OUTPUT_FIELDS
    ):
        errors.append(
            "OUTPUT_FIELDS_MISMATCH"
        )

    for field_name in [
        "question",
        "expected_answer",
        "question_intent",
    ]:

        value = parsed_json.get(
            field_name
        )

        if not isinstance(
            value,
            str,
        ):

            errors.append(
                f"{field_name.upper()}_NOT_STRING"
            )

        elif not value.strip():

            errors.append(
                f"{field_name.upper()}_EMPTY"
            )

    return sorted(
        set(errors)
    )


# =========================================================
# EXACT TWO-CANDIDATE REPAIR SET
# =========================================================

cell41_repair_candidate_ids = [
    audit["candidate_id"]
    for audit in cell40_failure_audit
]

assert len(
    cell41_repair_candidate_ids
) == 2

assert len(
    set(
        cell41_repair_candidate_ids
    )
) == 2

assert all(
    candidate_id
    in cell41_skeleton_by_candidate_id
    for candidate_id
    in cell41_repair_candidate_ids
)


cell41_repair_requests_v4 = []

for candidate_id in cell41_repair_candidate_ids:

    user_prompt = (
        cell41_build_generation_user_prompt_v4(
            candidate_id
        )
    )

    request_contract = {
        "candidate_id": candidate_id,

        "protocol_version": (
            GENERATION_V4_VERSION
        ),

        "protocol_sha256": (
            GENERATION_V4_PROTOCOL_SHA256
        ),

        "prompt_version": (
            GENERATION_V4_PROMPT_VERSION
        ),

        "output_schema_version": (
            GENERATION_V4_OUTPUT_SCHEMA_VERSION
        ),

        "model": (
            semantic_review_model
        ),

        "system_prompt": (
            GOLDEN_V2_GENERATION_SYSTEM_PROMPT_V4
        ),

        "user_prompt": (
            user_prompt
        ),
    }

    request_fingerprint = (
        cell41_sha256_text(
            cell41_canonical_json(
                request_contract
            )
        )
    )

    cell41_repair_requests_v4.append({
        **request_contract,

        "request_fingerprint": (
            request_fingerprint
        ),
    })


assert len(
    cell41_repair_requests_v4
) == 2


# =========================================================
# VERIFY REPAIR TARGETS MATCH OBSERVED FAILURES
# =========================================================

cell41_failure_dimensions = sorted({
    dimension
    for audit in cell40_failure_audit
    for dimension
    in audit["failed_dimensions"]
})

cell41_failure_classes = sorted({
    failure_class
    for audit in cell40_failure_audit
    for failure_class
    in audit["failure_classes"]
})


assert (
    "question_targets_claim"
    in cell41_failure_dimensions
)

assert (
    "answer_question_alignment"
    in cell41_failure_dimensions
)

assert (
    "modality_quantifier_fidelity"
    in cell41_failure_dimensions
)

assert (
    "QUESTION_TARGET_OR_ALIGNMENT_DRIFT"
    in cell41_failure_classes
)

assert (
    "CLAIM_FIDELITY_DRIFT"
    in cell41_failure_classes
)


# =========================================================
# SAFETY / ARCHITECTURE CHECKS
# =========================================================

cell41_checks = {
    "target_population_3813": (
        len(
            cell25_candidate_skeletons
        )
        == 3813
    ),

    "exactly_2_semantic_failures": (
        len(
            cell40_failure_audit
        )
        == 2
    ),

    "exactly_2_repair_requests": (
        len(
            cell41_repair_requests_v4
        )
        == 2
    ),

    "repair_targets_unique": (
        len(
            set(
                cell41_repair_candidate_ids
            )
        )
        == 2
    ),

    "question_target_failure_addressed": (
        "question_targets_claim"
        in cell41_failure_dimensions
    ),

    "answer_alignment_failure_addressed": (
        "answer_question_alignment"
        in cell41_failure_dimensions
    ),

    "quantifier_fidelity_failure_addressed": (
        "modality_quantifier_fidelity"
        in cell41_failure_dimensions
    ),

    "causal_question_guard_added": (
        cell41_protocol_contract_v4[
            "causal_question_requires_causal_claim"
        ]
        is True
    ),

    "qualifier_preservation_guard_added": (
        cell41_protocol_contract_v4[
            "meaning_bearing_qualifier_preservation_required"
        ]
        is True
    ),

    "qa_generation_only": (
        cell41_protocol_contract_v4[
            "generation_scope"
        ]
        == "QUESTION_EXPECTED_ANSWER_ONLY"
    ),

    "no_claim_evidence_revalidation": (
        cell41_protocol_contract_v4[
            "claim_evidence_revalidation"
        ]
        is False
    ),

    "no_support_verdict_generation": (
        cell41_protocol_contract_v4[
            "support_verdict_generation"
        ]
        is False
    ),

    "no_material_assertion_generation": (
        cell41_protocol_contract_v4[
            "material_assertion_generation"
        ]
        is False
    ),

    "three_output_fields_only": (
        GENERATION_V4_OUTPUT_FIELDS
        == {
            "question",
            "expected_answer",
            "question_intent",
        }
    ),

    "protocol_hash_created": (
        isinstance(
            GENERATION_V4_PROTOCOL_SHA256,
            str,
        )
        and len(
            GENERATION_V4_PROTOCOL_SHA256
        )
        == 64
    ),

    "zero_api_calls": True,

    "zero_file_writes": True,

    "full_generation_not_started": True,
}


cell41_status = (
    "PASS"
    if all(
        cell41_checks.values()
    )
    else "FAIL"
)


# =========================================================
# OUTPUT
# =========================================================

print(
    "=== GOLDEN V2 QA GENERATION V4 PROTOCOL REPAIR ==="
)

print()

print(
    "Protocol version:",
    GENERATION_V4_VERSION,
)

print(
    "Previous protocol:",
    GENERATION_V3_VERSION,
)

print(
    "Target population:",
    len(
        cell25_candidate_skeletons
    ),
)

print(
    "Observed semantic failures:",
    len(
        cell40_failure_audit
    ),
)

print(
    "Targeted repair population:",
    len(
        cell41_repair_requests_v4
    ),
)

print(
    "Repair candidate IDs:",
    cell41_repair_candidate_ids,
)

print(
    "Observed failed dimensions:",
    cell41_failure_dimensions,
)

print(
    "Observed failure classes:",
    cell41_failure_classes,
)

print(
    "Protocol SHA-256:",
    GENERATION_V4_PROTOCOL_SHA256,
)

print()

print(
    "Protocol changes:"
)

print(
    " - causal question requires causal claim: PASS"
)

print(
    " - question type must follow claim: PASS"
)

print(
    " - meaning-bearing qualifiers must be preserved: PASS"
)

print(
    " - negation/quantifiers/modality/numbers/comparisons preserved: PASS"
)

print(
    " - no Claim-Evidence revalidation: PASS"
)

print()

print(
    "Validation checks:"
)

for (
    check_name,
    check_value,
) in cell41_checks.items():

    print(
        " -",
        check_name,
        ":",
        (
            "PASS"
            if check_value
            else "FAIL"
        ),
    )

print()

print(
    "API calls performed:",
    0,
)

print(
    "Files modified:",
    0,
)

print(
    "Full Generation started:",
    False,
)

print(
    "CELL 41 STATUS:",
    cell41_status,
)

print(
    "Next readiness:",
    (
        "READY_FOR_TARGETED_V4_CANARY_REPAIR"
        if cell41_status == "PASS"
        else "BLOCKED_CELL41_REPAIR_REQUIRED"
    ),
)

assert cell41_status == "PASS"
````

در این مرحله فقط دو موردی که در Semantic Canary رد شدند با قرارداد اصلاح‌شده Generation v4 دوباره تولید می‌شوند.
ده خروجی سالم قبلی بدون تغییر حفظ می‌شوند.
هر درخواست دارای Fingerprint دقیق و Checkpoint مستقل است تا اجرای مجدد باعث فراخوانی دوباره مدل نشود.
این مرحله فقط بازتولید هدفمند دو مورد را انجام می‌دهد و Full Generation برای ۳۸۱۳ مورد هنوز آغاز نمی‌شود.

````python
# Code Cell 42 — Targeted Generation v4 Canary Repair

from collections import Counter
from pathlib import Path
import hashlib
import json
import os
import re
import tempfile
import time


# =========================================================
# CONFIG
# =========================================================

CELL42_RUNNER_VERSION = (
    "golden_v2_generation_v4_targeted_canary_repair_runner_v1"
)

CELL42_MAX_TOKENS = 1536
CELL42_TEMPERATURE = 0.7
CELL42_TOP_P = 0.8

CELL42_MAX_ATTEMPTS = 2
CELL42_RETRY_SECONDS = 2.0
CELL42_REQUEST_SPACING_SECONDS = 2.7

# Intentional paid execution for ONLY the two failed Canary cases.
# After successful execution this MUST become False before final Run All.
CELL42_ALLOW_API_FOR_MISSING = False


# =========================================================
# REQUIRED STATE
# =========================================================

required_cell42_names = [
    "project_root",
    "semantic_review_client",
    "semantic_review_model",
    "cell41_repair_requests_v4",
    "cell41_repair_candidate_ids",
    "cell41_build_generation_user_prompt_v4",
    "cell41_validate_generation_output_v4",
    "GOLDEN_V2_GENERATION_SYSTEM_PROMPT_V4",
    "GENERATION_V4_VERSION",
    "GENERATION_V4_PROMPT_VERSION",
    "GENERATION_V4_OUTPUT_SCHEMA_VERSION",
    "GENERATION_V4_PROTOCOL_SHA256",
    "cell36_final_canary_results",
]

missing_cell42_names = [
    name
    for name in required_cell42_names
    if name not in globals()
]

assert not missing_cell42_names, (
    "Missing required Cell 42 state: "
    + ", ".join(missing_cell42_names)
)

assert len(cell41_repair_requests_v4) == 2
assert len(cell41_repair_candidate_ids) == 2
assert len(set(cell41_repair_candidate_ids)) == 2

assert len(cell36_final_canary_results) == 12
assert len({
    result["candidate_id"]
    for result in cell36_final_canary_results
}) == 12

assert set(cell41_repair_candidate_ids) == {
    request["candidate_id"]
    for request in cell41_repair_requests_v4
}


# =========================================================
# HASH UTILITIES
# =========================================================

def cell42_canonical_json(value):

    return json.dumps(
        value,
        ensure_ascii=False,
        sort_keys=True,
        separators=(",", ":"),
    )


def cell42_sha256_text(text):

    return hashlib.sha256(
        text.encode("utf-8")
    ).hexdigest()


# =========================================================
# RESPONSE CLEANING
# =========================================================

def cell42_clean_model_json(content):

    if not isinstance(content, str):
        return ""

    clean = content.strip()

    if clean.startswith("```"):

        clean = re.sub(
            r"^```(?:json)?\s*",
            "",
            clean,
            flags=re.IGNORECASE,
        )

        clean = re.sub(
            r"\s*```$",
            "",
            clean,
        )

    return clean.strip()


def cell42_is_rate_limit_exception(exc):

    error_text = str(exc).lower()
    error_type = type(exc).__name__.lower()

    return (
        "ratelimit" in error_type
        or "rate_limit" in error_type
        or "429" in error_text
        or "too many requests" in error_text
    )


# =========================================================
# EXACT REQUEST CONTRACT
# =========================================================

def cell42_build_request(
    source_request,
):

    candidate_id = source_request[
        "candidate_id"
    ]

    user_prompt = (
        cell41_build_generation_user_prompt_v4(
            candidate_id
        )
    )

    request_contract = {
        "runner_version": (
            CELL42_RUNNER_VERSION
        ),

        "candidate_id": (
            candidate_id
        ),

        "protocol_version": (
            GENERATION_V4_VERSION
        ),

        "protocol_sha256": (
            GENERATION_V4_PROTOCOL_SHA256
        ),

        "prompt_version": (
            GENERATION_V4_PROMPT_VERSION
        ),

        "output_schema_version": (
            GENERATION_V4_OUTPUT_SCHEMA_VERSION
        ),

        "model": (
            semantic_review_model
        ),

        "system_prompt": (
            GOLDEN_V2_GENERATION_SYSTEM_PROMPT_V4
        ),

        "user_prompt": (
            user_prompt
        ),

        "temperature": (
            CELL42_TEMPERATURE
        ),

        "top_p": (
            CELL42_TOP_P
        ),

        "max_tokens": (
            CELL42_MAX_TOKENS
        ),

        "enable_thinking": False,
    }

    request_fingerprint = (
        cell42_sha256_text(
            cell42_canonical_json(
                request_contract
            )
        )
    )

    return {
        **request_contract,

        "request_fingerprint": (
            request_fingerprint
        ),
    }


cell42_requests = [
    cell42_build_request(
        source_request
    )
    for source_request
    in cell41_repair_requests_v4
]

assert len(cell42_requests) == 2

assert {
    request["candidate_id"]
    for request in cell42_requests
} == set(
    cell41_repair_candidate_ids
)


# =========================================================
# CHECKPOINT DIRECTORY
# =========================================================

cell42_checkpoint_dir = (
    Path(project_root)
    / "artifacts"
    / "phase2"
    / "golden_dataset_v2"
    / "work"
    / "generation_v4_targeted_canary_repair"
)

cell42_checkpoint_dir.mkdir(
    parents=True,
    exist_ok=True,
)


def cell42_checkpoint_path(
    candidate_id,
):

    safe_id = re.sub(
        r"[^A-Za-z0-9_.-]+",
        "_",
        candidate_id,
    )

    return (
        cell42_checkpoint_dir
        / f"{safe_id}.json"
    )


# =========================================================
# ATOMIC WRITE
# =========================================================

def cell42_atomic_write_json(
    path,
    payload,
):

    path = Path(path)

    content = (
        json.dumps(
            payload,
            ensure_ascii=False,
            sort_keys=True,
            indent=2,
        )
        + "\n"
    )

    fd, temp_name = tempfile.mkstemp(
        prefix=path.name + ".",
        suffix=".tmp",
        dir=str(path.parent),
    )

    try:

        with os.fdopen(
            fd,
            "w",
            encoding="utf-8",
        ) as handle:

            handle.write(content)
            handle.flush()
            os.fsync(
                handle.fileno()
            )

        os.replace(
            temp_name,
            path,
        )

    except Exception:

        try:
            os.unlink(temp_name)
        except FileNotFoundError:
            pass

        raise


# =========================================================
# CHECKPOINT CONTRACT
# =========================================================

def cell42_build_checkpoint(
    request,
    result,
):

    checkpoint = {
        "checkpoint_version": (
            CELL42_RUNNER_VERSION
        ),

        "candidate_id": (
            request["candidate_id"]
        ),

        "request_fingerprint": (
            request[
                "request_fingerprint"
            ]
        ),

        "protocol_version": (
            GENERATION_V4_VERSION
        ),

        "protocol_sha256": (
            GENERATION_V4_PROTOCOL_SHA256
        ),

        "prompt_version": (
            GENERATION_V4_PROMPT_VERSION
        ),

        "output_schema_version": (
            GENERATION_V4_OUTPUT_SCHEMA_VERSION
        ),

        "model": (
            semantic_review_model
        ),

        "max_tokens": (
            CELL42_MAX_TOKENS
        ),

        "result": result,
    }

    checkpoint[
        "checkpoint_payload_sha256"
    ] = cell42_sha256_text(
        cell42_canonical_json(
            checkpoint
        )
    )

    return checkpoint


def cell42_validate_checkpoint(
    checkpoint,
    request,
):

    errors = []

    if not isinstance(
        checkpoint,
        dict,
    ):
        return [
            "CHECKPOINT_NOT_OBJECT"
        ]

    stored_sha = checkpoint.get(
        "checkpoint_payload_sha256"
    )

    payload_without_sha = {
        key: value
        for key, value
        in checkpoint.items()
        if key
        != "checkpoint_payload_sha256"
    }

    recalculated_sha = (
        cell42_sha256_text(
            cell42_canonical_json(
                payload_without_sha
            )
        )
    )

    if stored_sha != recalculated_sha:

        errors.append(
            "CHECKPOINT_HASH_MISMATCH"
        )

    exact_expectations = {
        "checkpoint_version": (
            CELL42_RUNNER_VERSION
        ),

        "candidate_id": (
            request["candidate_id"]
        ),

        "request_fingerprint": (
            request[
                "request_fingerprint"
            ]
        ),

        "protocol_version": (
            GENERATION_V4_VERSION
        ),

        "protocol_sha256": (
            GENERATION_V4_PROTOCOL_SHA256
        ),

        "prompt_version": (
            GENERATION_V4_PROMPT_VERSION
        ),

        "output_schema_version": (
            GENERATION_V4_OUTPUT_SCHEMA_VERSION
        ),

        "model": (
            semantic_review_model
        ),

        "max_tokens": (
            CELL42_MAX_TOKENS
        ),
    }

    for (
        field_name,
        expected_value,
    ) in exact_expectations.items():

        if (
            checkpoint.get(
                field_name
            )
            != expected_value
        ):

            errors.append(
                f"{field_name.upper()}_MISMATCH"
            )

    result = checkpoint.get(
        "result"
    )

    if not isinstance(
        result,
        dict,
    ):

        errors.append(
            "CHECKPOINT_RESULT_INVALID"
        )

    elif (
        result.get(
            "candidate_id"
        )
        != request[
            "candidate_id"
        ]
    ):

        errors.append(
            "CHECKPOINT_RESULT_CANDIDATE_MISMATCH"
        )

    return sorted(
        set(errors)
    )


def cell42_load_checkpoint(
    request,
):

    path = cell42_checkpoint_path(
        request[
            "candidate_id"
        ]
    )

    if not path.exists():

        return (
            None,
            "MISSING",
            [],
        )

    try:

        with path.open(
            "r",
            encoding="utf-8",
        ) as handle:

            checkpoint = json.load(
                handle
            )

    except Exception as exc:

        return (
            None,
            "UNREADABLE",
            [
                f"{type(exc).__name__}: {exc}"
            ],
        )

    errors = (
        cell42_validate_checkpoint(
            checkpoint,
            request,
        )
    )

    if errors:

        return (
            checkpoint,
            "INVALID",
            errors,
        )

    return (
        checkpoint,
        "VALID",
        [],
    )


# =========================================================
# COST
# =========================================================

cell42_has_cost_rates = (
    "INPUT_TOMAN_PER_TOKEN" in globals()
    and
    "OUTPUT_TOMAN_PER_TOKEN" in globals()
)


def cell42_estimated_cost_toman(
    input_tokens,
    output_tokens,
):

    if not cell42_has_cost_rates:
        return None

    return (
        input_tokens
        * INPUT_TOMAN_PER_TOKEN
        +
        output_tokens
        * OUTPUT_TOMAN_PER_TOKEN
    )


# =========================================================
# API EXECUTION
# =========================================================

def cell42_execute_request(
    request,
):

    attempt_history = []

    for attempt_number in range(
        1,
        CELL42_MAX_ATTEMPTS + 1,
    ):

        started_at = time.time()

        try:

            response = (
                semantic_review_client
                .chat
                .completions
                .create(
                    model=(
                        semantic_review_model
                    ),

                    messages=[
                        {
                            "role": "system",
                            "content": (
                                GOLDEN_V2_GENERATION_SYSTEM_PROMPT_V4
                            ),
                        },
                        {
                            "role": "user",
                            "content": (
                                request[
                                    "user_prompt"
                                ]
                            ),
                        },
                    ],

                    temperature=(
                        CELL42_TEMPERATURE
                    ),

                    top_p=(
                        CELL42_TOP_P
                    ),

                    max_tokens=(
                        CELL42_MAX_TOKENS
                    ),

                    extra_body={
                        "enable_thinking": False,
                    },
                )
            )

        except Exception as exc:

            attempt_history.append({
                "attempt": (
                    attempt_number
                ),

                "status": (
                    "API_ERROR"
                ),

                "error": (
                    f"{type(exc).__name__}: {exc}"
                ),
            })

            if (
                attempt_number
                < CELL42_MAX_ATTEMPTS
            ):

                wait_seconds = (
                    70.0
                    if cell42_is_rate_limit_exception(
                        exc
                    )
                    else (
                        CELL42_RETRY_SECONDS
                        * attempt_number
                    )
                )

                time.sleep(
                    wait_seconds
                )

                continue

            return {
                "candidate_id": (
                    request[
                        "candidate_id"
                    ]
                ),

                "execution_status": (
                    "API_FAILED"
                ),

                "attempt_history": (
                    attempt_history
                ),
            }

        elapsed_seconds = (
            time.time()
            - started_at
        )

        choice = response.choices[0]

        finish_reason = getattr(
            choice,
            "finish_reason",
            None,
        )

        raw_content = getattr(
            choice.message,
            "content",
            None,
        )

        usage = getattr(
            response,
            "usage",
            None,
        )

        input_tokens = int(
            getattr(
                usage,
                "prompt_tokens",
                0,
            )
            or 0
        )

        output_tokens = int(
            getattr(
                usage,
                "completion_tokens",
                0,
            )
            or 0
        )

        total_tokens = int(
            getattr(
                usage,
                "total_tokens",
                0,
            )
            or (
                input_tokens
                + output_tokens
            )
        )

        reasoning_tokens = 0

        completion_details = getattr(
            usage,
            "completion_tokens_details",
            None,
        )

        if completion_details is not None:

            reasoning_tokens = int(
                getattr(
                    completion_details,
                    "reasoning_tokens",
                    0,
                )
                or 0
            )

        clean_content = (
            cell42_clean_model_json(
                raw_content
            )
        )

        parsed_json = None
        parse_error = None

        try:

            parsed_json = json.loads(
                clean_content
            )

        except Exception as exc:

            parse_error = (
                f"{type(exc).__name__}: {exc}"
            )

        if parsed_json is None:

            structural_errors = [
                "JSON_PARSE_FAILED"
            ]

        else:

            structural_errors = (
                cell41_validate_generation_output_v4(
                    parsed_json
                )
            )

        return {
            "candidate_id": (
                request[
                    "candidate_id"
                ]
            ),

            "request_fingerprint": (
                request[
                    "request_fingerprint"
                ]
            ),

            "protocol_version": (
                GENERATION_V4_VERSION
            ),

            "protocol_sha256": (
                GENERATION_V4_PROTOCOL_SHA256
            ),

            "model": (
                semantic_review_model
            ),

            "finish_reason": (
                finish_reason
            ),

            "elapsed_seconds": round(
                elapsed_seconds,
                4,
            ),

            "usage": {
                "input_tokens": (
                    input_tokens
                ),

                "output_tokens": (
                    output_tokens
                ),

                "reasoning_tokens": (
                    reasoning_tokens
                ),

                "total_tokens": (
                    total_tokens
                ),
            },

            "estimated_cost_toman": (
                cell42_estimated_cost_toman(
                    input_tokens,
                    output_tokens,
                )
            ),

            "raw_content": (
                raw_content
            ),

            "parsed_json": (
                parsed_json
            ),

            "parse_error": (
                parse_error
            ),

            "structural_errors": (
                structural_errors
            ),

            "structural_status": (
                "PASS"
                if not structural_errors
                else "FAIL"
            ),

            "attempt_history": (
                attempt_history
            ),
        }

    raise RuntimeError(
        "Unexpected Cell 42 execution path."
    )


# =========================================================
# PREFLIGHT
# =========================================================

print(
    "=== GOLDEN V2 TARGETED GENERATION V4 REPAIR PREFLIGHT ==="
)

print()

print(
    "Repair population:",
    len(cell42_requests),
)

print(
    "Repair candidates:",
    [
        request["candidate_id"]
        for request in cell42_requests
    ],
)

print(
    "Model:",
    semantic_review_model,
)

print(
    "Protocol:",
    GENERATION_V4_VERSION,
)

print(
    "Protocol SHA:",
    GENERATION_V4_PROTOCOL_SHA256,
)

print(
    "Max tokens:",
    CELL42_MAX_TOKENS,
)

print(
    "API allowed for missing exact checkpoints:",
    CELL42_ALLOW_API_FOR_MISSING,
)

print(
    "Full Generation started:",
    False,
)

print()


# =========================================================
# RESOLVE EXACT TWO TARGETS
# =========================================================

cell42_repair_results = []
cell42_resolution_modes = []

cell42_api_calls_this_execution = 0
cell42_checkpoints_created_this_execution = 0


for index, request in enumerate(
    cell42_requests,
    start=1,
):

    (
        checkpoint,
        checkpoint_status,
        checkpoint_errors,
    ) = cell42_load_checkpoint(
        request
    )

    candidate_id = request[
        "candidate_id"
    ]

    if checkpoint_status == "VALID":

        result = checkpoint[
            "result"
        ]

        resolution_mode = (
            "EXACT_CHECKPOINT"
        )

    else:

        if checkpoint_status in {
            "INVALID",
            "UNREADABLE",
        }:

            raise AssertionError(
                "Existing Cell 42 checkpoint is invalid. "
                "Refusing automatic paid replacement for "
                f"{candidate_id}. "
                f"Status={checkpoint_status}; "
                f"errors={checkpoint_errors}"
            )

        if not CELL42_ALLOW_API_FOR_MISSING:

            raise AssertionError(
                "Exact Cell 42 checkpoint is missing and "
                "API execution is disabled for "
                f"{candidate_id}."
            )

        if (
            cell42_api_calls_this_execution
            > 0
        ):

            time.sleep(
                CELL42_REQUEST_SPACING_SECONDS
            )

        result = (
            cell42_execute_request(
                request
            )
        )

        cell42_api_calls_this_execution += 1

        checkpoint = (
            cell42_build_checkpoint(
                request,
                result,
            )
        )

        cell42_atomic_write_json(
            cell42_checkpoint_path(
                candidate_id
            ),
            checkpoint,
        )

        (
            verified_checkpoint,
            verified_status,
            verified_errors,
        ) = cell42_load_checkpoint(
            request
        )

        assert (
            verified_status
            == "VALID"
        ), (
            "New Cell 42 checkpoint verification failed "
            f"for {candidate_id}: "
            f"{verified_errors}"
        )

        result = (
            verified_checkpoint[
                "result"
            ]
        )

        resolution_mode = (
            "NEW_PAID_EXECUTION"
        )

        cell42_checkpoints_created_this_execution += 1

    cell42_repair_results.append(
        result
    )

    cell42_resolution_modes.append({
        "candidate_id": (
            candidate_id
        ),

        "resolution_mode": (
            resolution_mode
        ),
    })

    print(
        f"[{index:02d}/02]",
        candidate_id,
        "|",
        resolution_mode,
        "| structural=",
        result.get(
            "structural_status",
            result.get(
                "execution_status"
            ),
        ),
        "| finish_reason=",
        result.get(
            "finish_reason"
        ),
        flush=True,
    )


# =========================================================
# FINAL DISK RESCAN
# =========================================================

cell42_valid_checkpoint_count = 0
cell42_missing_checkpoint_count = 0
cell42_invalid_checkpoint_count = 0


for request in cell42_requests:

    (
        checkpoint,
        status,
        errors,
    ) = cell42_load_checkpoint(
        request
    )

    if status == "VALID":

        cell42_valid_checkpoint_count += 1

    elif status == "MISSING":

        cell42_missing_checkpoint_count += 1

    else:

        cell42_invalid_checkpoint_count += 1


# =========================================================
# RESULT COUNTS
# =========================================================

cell42_structural_pass_count = sum(
    result.get(
        "structural_status"
    )
    == "PASS"
    for result in cell42_repair_results
)

cell42_api_failure_count = sum(
    result.get(
        "execution_status"
    )
    == "API_FAILED"
    for result in cell42_repair_results
)

cell42_finish_reason_distribution = Counter(
    result.get(
        "finish_reason"
    )
    for result in cell42_repair_results
)

cell42_known_costs = [
    result.get(
        "estimated_cost_toman"
    )
    for result in cell42_repair_results
    if isinstance(
        result.get(
            "estimated_cost_toman"
        ),
        (int, float),
    )
]

cell42_total_estimated_cost_toman = (
    sum(cell42_known_costs)
    if cell42_known_costs
    else None
)


# =========================================================
# RECONSTRUCT 12-CANDIDATE GENERATION SET
# =========================================================

cell42_repair_by_candidate_id = {
    result["candidate_id"]: result
    for result in cell42_repair_results
}

assert len(
    cell42_repair_by_candidate_id
) == 2


cell42_final_canary_generation_results = []

for historical_result in cell36_final_canary_results:

    candidate_id = (
        historical_result[
            "candidate_id"
        ]
    )

    if (
        candidate_id
        in cell42_repair_by_candidate_id
    ):

        cell42_final_canary_generation_results.append(
            cell42_repair_by_candidate_id[
                candidate_id
            ]
        )

    else:

        cell42_final_canary_generation_results.append(
            historical_result
        )


assert len(
    cell42_final_canary_generation_results
) == 12

assert len({
    result["candidate_id"]
    for result
    in cell42_final_canary_generation_results
}) == 12


# =========================================================
# RUNNER HEALTH
# =========================================================

cell42_runner_checks = {
    "exactly_2_repair_targets": (
        len(cell42_requests)
        == 2
    ),

    "2_results_reconstructed": (
        len(cell42_repair_results)
        == 2
    ),

    "2_valid_checkpoints": (
        cell42_valid_checkpoint_count
        == 2
    ),

    "zero_missing_checkpoints": (
        cell42_missing_checkpoint_count
        == 0
    ),

    "zero_invalid_checkpoints": (
        cell42_invalid_checkpoint_count
        == 0
    ),

    "zero_api_failures": (
        cell42_api_failure_count
        == 0
    ),

    "final_canary_population_12": (
        len(
            cell42_final_canary_generation_results
        )
        == 12
    ),

    "full_generation_not_started": True,
}


cell42_runner_status = (
    "PASS"
    if all(
        cell42_runner_checks.values()
    )
    else "FAIL"
)


cell42_experimental_status = (
    "PASS"
    if cell42_structural_pass_count
    == 2
    else "FAIL"
)


if cell42_runner_status != "PASS":

    cell42_next_readiness = (
        "BLOCKED_CELL42_RUNNER_REPAIR_REQUIRED"
    )

elif cell42_experimental_status == "PASS":

    cell42_next_readiness = (
        "READY_FOR_TARGETED_V4_SEMANTIC_VALIDATION"
    )

else:

    cell42_next_readiness = (
        "READY_FOR_CELL42_GENERATION_FAILURE_ANALYSIS"
    )


# =========================================================
# OUTPUT
# =========================================================

print()

print(
    "=== GOLDEN V2 TARGETED GENERATION V4 REPAIR RESULT ==="
)

print()

print(
    "Repair results reconstructed:",
    len(cell42_repair_results),
)

print(
    "Valid checkpoints:",
    cell42_valid_checkpoint_count,
)

print(
    "Missing checkpoints:",
    cell42_missing_checkpoint_count,
)

print(
    "Invalid checkpoints:",
    cell42_invalid_checkpoint_count,
)

print(
    "Resolution modes:",
    dict(
        Counter(
            row["resolution_mode"]
            for row
            in cell42_resolution_modes
        )
    ),
)

print(
    "API calls this execution:",
    cell42_api_calls_this_execution,
)

print(
    "Checkpoints created this execution:",
    cell42_checkpoints_created_this_execution,
)

print(
    "Structural PASS:",
    cell42_structural_pass_count,
    "/ 2",
)

print(
    "Finish reason distribution:",
    dict(
        cell42_finish_reason_distribution
    ),
)

print(
    "Estimated total cost (toman):",
    cell42_total_estimated_cost_toman,
)

print()

print(
    "Generated repairs:"
)

for result in cell42_repair_results:

    print()
    print(
        "Candidate:",
        result["candidate_id"],
    )

    print(
        json.dumps(
            result.get(
                "parsed_json"
            ),
            ensure_ascii=False,
            indent=2,
        )
    )


print()
print(
    "Runner checks:"
)

for (
    check_name,
    check_value,
) in cell42_runner_checks.items():

    print(
        " -",
        check_name,
        ":",
        (
            "PASS"
            if check_value
            else "FAIL"
        ),
    )


print()

print(
    "Runner health:",
    cell42_runner_status,
)

print(
    "Targeted Generation result:",
    cell42_experimental_status,
)

print(
    "Final Canary generation population:",
    len(
        cell42_final_canary_generation_results
    ),
)

print(
    "Full Generation started:",
    False,
)

print(
    "Next readiness:",
    cell42_next_readiness,
)

print(
    "CELL 42 CHECKPOINTED RUNNER:",
    cell42_runner_status,
)


assert cell42_runner_status == "PASS", (
    "Cell 42 checkpointed runner health failed."
)
````

در این مرحله فقط دو خروجی جدیدی که با Generation v4 بازتولید شدند، در برابر Claim و Evidence از پیش اعتبارسنجی‌شده بررسی می‌شوند.
ده مورد Semantic PASS قبلی بدون اجرای مجدد حفظ می‌شوند.
برای هر داوری Checkpoint و Fingerprint مستقل ساخته می‌شود تا اجرای مجدد باعث فراخوانی دوبارۀ مدل نشود.
Full Generation در این مرحله آغاز نمی‌شود.
</p>

````python
# Code Cell 43 — Targeted v4 Semantic Validation

from collections import Counter
from pathlib import Path
import hashlib
import json
import os
import re
import tempfile
import time


# =========================================================
# CONFIG
# =========================================================

CELL43_RUNNER_VERSION = (
    "golden_v2_qa_semantic_v4_targeted_validator_v1"
)

CELL43_VALIDATION_PROTOCOL_VERSION = (
    "golden_v2_qa_semantic_validation_v1"
)

CELL43_OUTPUT_SCHEMA_VERSION = (
    "golden_v2_qa_semantic_validation_output_v1"
)

CELL43_MAX_TOKENS = 1536
CELL43_TEMPERATURE = 0.0
CELL43_TOP_P = 1.0

CELL43_MAX_ATTEMPTS = 2
CELL43_RETRY_SECONDS = 2.0
CELL43_REQUEST_SPACING_SECONDS = 2.7

# Intentional paid validation of ONLY the two new v4 generations.
# After successful execution this MUST become False before final Run All.
CELL43_ALLOW_API_FOR_MISSING = False


# =========================================================
# REQUIRED STATE
# =========================================================

required_cell43_names = [
    "project_root",
    "semantic_review_client",
    "semantic_review_model",
    "cell42_repair_results",
    "cell41_repair_candidate_ids",
    "cell39_final_validation_results",
    "cell37_skeleton_by_candidate_id",
    "cell37_validate_output",
    "cell37_clean_model_json",
    "cell37_is_rate_limit_exception",
    "CELL37_SYSTEM_PROMPT",
    "CELL37_BOOLEAN_FIELDS",
    "GENERATION_V4_VERSION",
    "GENERATION_V4_PROTOCOL_SHA256",
]

missing_cell43_names = [
    name
    for name in required_cell43_names
    if name not in globals()
]

assert not missing_cell43_names, (
    "Missing required Cell 43 state: "
    + ", ".join(missing_cell43_names)
)

assert len(cell42_repair_results) == 2
assert len(cell41_repair_candidate_ids) == 2
assert len(set(cell41_repair_candidate_ids)) == 2

assert len(cell39_final_validation_results) == 12

assert all(
    result.get("structural_status") == "PASS"
    for result in cell42_repair_results
), (
    "Cell 43 requires 2/2 structurally valid v4 generations."
)

assert {
    result["candidate_id"]
    for result in cell42_repair_results
} == set(cell41_repair_candidate_ids)


# =========================================================
# LOOKUPS
# =========================================================

cell43_generation_by_candidate_id = {
    result["candidate_id"]: result
    for result in cell42_repair_results
}

assert len(cell43_generation_by_candidate_id) == 2

assert all(
    candidate_id in cell37_skeleton_by_candidate_id
    for candidate_id in cell41_repair_candidate_ids
)


# =========================================================
# HASH UTILITIES
# =========================================================

def cell43_canonical_json(value):

    return json.dumps(
        value,
        ensure_ascii=False,
        sort_keys=True,
        separators=(",", ":"),
    )


def cell43_sha256_text(text):

    return hashlib.sha256(
        text.encode("utf-8")
    ).hexdigest()


# =========================================================
# VALIDATION PAYLOAD
# =========================================================

def cell43_build_payload(candidate_id):

    skeleton = (
        cell37_skeleton_by_candidate_id[
            candidate_id
        ]
    )

    generation_result = (
        cell43_generation_by_candidate_id[
            candidate_id
        ]
    )

    generated = generation_result.get(
        "parsed_json"
    )

    assert isinstance(
        generated,
        dict,
    )

    evidence = []

    for (
        evidence_id,
        evidence_quote,
        evidence_locator,
    ) in zip(
        skeleton["evidence_ids"],
        skeleton["evidence_quotes"],
        skeleton["evidence_locators"],
    ):

        evidence.append({
            "evidence_id": evidence_id,
            "quote": evidence_quote,
            "locator": evidence_locator,
        })

    return {
        "candidate_id": candidate_id,

        "target_claim": {
            "claim_id": (
                skeleton["claim_id"]
            ),

            "text_fa": (
                skeleton["claim_text_fa"]
            ),

            "claim_type": (
                skeleton["claim_type"]
            ),

            "claim_types": (
                skeleton["claim_types"]
            ),

            "authority_class": (
                skeleton["authority_class"]
            ),
        },

        "evidence": evidence,

        "generated_qa": {
            "question": (
                generated["question"]
            ),

            "expected_answer": (
                generated["expected_answer"]
            ),

            "question_intent": (
                generated["question_intent"]
            ),
        },
    }


def cell43_build_user_prompt(candidate_id):

    payload = cell43_build_payload(
        candidate_id
    )

    return (
        "Validate only the generated Question and Expected Answer "
        "against the pre-validated target Claim and Evidence.\n\n"
        + json.dumps(
            payload,
            ensure_ascii=False,
            sort_keys=True,
            indent=2,
        )
    )


# =========================================================
# EXACT REQUEST CONTRACT
# =========================================================

def cell43_build_request(candidate_id):

    generation_result = (
        cell43_generation_by_candidate_id[
            candidate_id
        ]
    )

    generated_qa_sha256 = (
        cell43_sha256_text(
            cell43_canonical_json(
                generation_result[
                    "parsed_json"
                ]
            )
        )
    )

    user_prompt = (
        cell43_build_user_prompt(
            candidate_id
        )
    )

    fingerprint_payload = {
        "runner_version": (
            CELL43_RUNNER_VERSION
        ),

        "validation_protocol_version": (
            CELL43_VALIDATION_PROTOCOL_VERSION
        ),

        "output_schema_version": (
            CELL43_OUTPUT_SCHEMA_VERSION
        ),

        "candidate_id": candidate_id,

        "generation_protocol_version": (
            GENERATION_V4_VERSION
        ),

        "generation_protocol_sha256": (
            GENERATION_V4_PROTOCOL_SHA256
        ),

        "generated_qa_sha256": (
            generated_qa_sha256
        ),

        "model": (
            semantic_review_model
        ),

        "system_prompt": (
            CELL37_SYSTEM_PROMPT
        ),

        "user_prompt": (
            user_prompt
        ),

        "temperature": (
            CELL43_TEMPERATURE
        ),

        "top_p": (
            CELL43_TOP_P
        ),

        "max_tokens": (
            CELL43_MAX_TOKENS
        ),

        "enable_thinking": False,
    }

    request_fingerprint = (
        cell43_sha256_text(
            cell43_canonical_json(
                fingerprint_payload
            )
        )
    )

    return {
        "candidate_id": candidate_id,

        "user_prompt": user_prompt,

        "generated_qa_sha256": (
            generated_qa_sha256
        ),

        "request_fingerprint": (
            request_fingerprint
        ),
    }


cell43_requests = [
    cell43_build_request(
        candidate_id
    )
    for candidate_id
    in cell41_repair_candidate_ids
]

assert len(cell43_requests) == 2


# =========================================================
# CHECKPOINT DIRECTORY
# =========================================================

cell43_checkpoint_dir = (
    Path(project_root)
    / "artifacts"
    / "phase2"
    / "golden_dataset_v2"
    / "work"
    / "qa_semantic_v4_targeted_validation"
)

cell43_checkpoint_dir.mkdir(
    parents=True,
    exist_ok=True,
)


def cell43_checkpoint_path(candidate_id):

    safe_id = re.sub(
        r"[^A-Za-z0-9_.-]+",
        "_",
        candidate_id,
    )

    return (
        cell43_checkpoint_dir
        / f"{safe_id}.json"
    )


# =========================================================
# ATOMIC WRITE
# =========================================================

def cell43_atomic_write_json(
    path,
    payload,
):

    path = Path(path)

    content = (
        json.dumps(
            payload,
            ensure_ascii=False,
            sort_keys=True,
            indent=2,
        )
        + "\n"
    )

    fd, temp_name = tempfile.mkstemp(
        prefix=path.name + ".",
        suffix=".tmp",
        dir=str(path.parent),
    )

    try:

        with os.fdopen(
            fd,
            "w",
            encoding="utf-8",
        ) as handle:

            handle.write(content)
            handle.flush()
            os.fsync(
                handle.fileno()
            )

        os.replace(
            temp_name,
            path,
        )

    except Exception:

        try:
            os.unlink(
                temp_name
            )
        except FileNotFoundError:
            pass

        raise


# =========================================================
# CHECKPOINT CONTRACT
# =========================================================

def cell43_build_checkpoint(
    request,
    result,
):

    checkpoint = {
        "checkpoint_version": (
            CELL43_RUNNER_VERSION
        ),

        "candidate_id": (
            request["candidate_id"]
        ),

        "request_fingerprint": (
            request[
                "request_fingerprint"
            ]
        ),

        "generated_qa_sha256": (
            request[
                "generated_qa_sha256"
            ]
        ),

        "validation_protocol_version": (
            CELL43_VALIDATION_PROTOCOL_VERSION
        ),

        "output_schema_version": (
            CELL43_OUTPUT_SCHEMA_VERSION
        ),

        "generation_protocol_version": (
            GENERATION_V4_VERSION
        ),

        "generation_protocol_sha256": (
            GENERATION_V4_PROTOCOL_SHA256
        ),

        "model": (
            semantic_review_model
        ),

        "max_tokens": (
            CELL43_MAX_TOKENS
        ),

        "result": result,
    }

    checkpoint[
        "checkpoint_payload_sha256"
    ] = cell43_sha256_text(
        cell43_canonical_json(
            checkpoint
        )
    )

    return checkpoint


def cell43_validate_checkpoint(
    checkpoint,
    request,
):

    errors = []

    if not isinstance(
        checkpoint,
        dict,
    ):
        return [
            "CHECKPOINT_NOT_OBJECT"
        ]

    stored_sha = checkpoint.get(
        "checkpoint_payload_sha256"
    )

    payload_without_sha = {
        key: value
        for key, value
        in checkpoint.items()
        if key
        != "checkpoint_payload_sha256"
    }

    recalculated_sha = (
        cell43_sha256_text(
            cell43_canonical_json(
                payload_without_sha
            )
        )
    )

    if stored_sha != recalculated_sha:

        errors.append(
            "CHECKPOINT_HASH_MISMATCH"
        )

    exact_expectations = {
        "checkpoint_version": (
            CELL43_RUNNER_VERSION
        ),

        "candidate_id": (
            request["candidate_id"]
        ),

        "request_fingerprint": (
            request[
                "request_fingerprint"
            ]
        ),

        "generated_qa_sha256": (
            request[
                "generated_qa_sha256"
            ]
        ),

        "validation_protocol_version": (
            CELL43_VALIDATION_PROTOCOL_VERSION
        ),

        "output_schema_version": (
            CELL43_OUTPUT_SCHEMA_VERSION
        ),

        "generation_protocol_version": (
            GENERATION_V4_VERSION
        ),

        "generation_protocol_sha256": (
            GENERATION_V4_PROTOCOL_SHA256
        ),

        "model": (
            semantic_review_model
        ),

        "max_tokens": (
            CELL43_MAX_TOKENS
        ),
    }

    for (
        field_name,
        expected_value,
    ) in exact_expectations.items():

        if (
            checkpoint.get(
                field_name
            )
            != expected_value
        ):

            errors.append(
                f"{field_name.upper()}_MISMATCH"
            )

    result = checkpoint.get(
        "result"
    )

    if not isinstance(
        result,
        dict,
    ):

        errors.append(
            "CHECKPOINT_RESULT_INVALID"
        )

    elif (
        result.get(
            "candidate_id"
        )
        != request[
            "candidate_id"
        ]
    ):

        errors.append(
            "CHECKPOINT_RESULT_CANDIDATE_MISMATCH"
        )

    return sorted(
        set(errors)
    )


def cell43_load_checkpoint(request):

    path = cell43_checkpoint_path(
        request["candidate_id"]
    )

    if not path.exists():

        return (
            None,
            "MISSING",
            [],
        )

    try:

        with path.open(
            "r",
            encoding="utf-8",
        ) as handle:

            checkpoint = json.load(
                handle
            )

    except Exception as exc:

        return (
            None,
            "UNREADABLE",
            [
                f"{type(exc).__name__}: {exc}"
            ],
        )

    errors = (
        cell43_validate_checkpoint(
            checkpoint,
            request,
        )
    )

    if errors:

        return (
            checkpoint,
            "INVALID",
            errors,
        )

    return (
        checkpoint,
        "VALID",
        [],
    )


# =========================================================
# COST
# =========================================================

cell43_has_cost_rates = (
    "INPUT_TOMAN_PER_TOKEN" in globals()
    and
    "OUTPUT_TOMAN_PER_TOKEN" in globals()
)


def cell43_estimated_cost_toman(
    input_tokens,
    output_tokens,
):

    if not cell43_has_cost_rates:
        return None

    return (
        input_tokens
        * INPUT_TOMAN_PER_TOKEN
        +
        output_tokens
        * OUTPUT_TOMAN_PER_TOKEN
    )


# =========================================================
# API EXECUTION
# =========================================================

def cell43_execute_request(request):

    attempt_history = []

    for attempt_number in range(
        1,
        CELL43_MAX_ATTEMPTS + 1,
    ):

        started_at = time.time()

        try:

            response = (
                semantic_review_client
                .chat
                .completions
                .create(
                    model=(
                        semantic_review_model
                    ),

                    messages=[
                        {
                            "role": "system",
                            "content": (
                                CELL37_SYSTEM_PROMPT
                            ),
                        },
                        {
                            "role": "user",
                            "content": (
                                request[
                                    "user_prompt"
                                ]
                            ),
                        },
                    ],

                    temperature=(
                        CELL43_TEMPERATURE
                    ),

                    top_p=(
                        CELL43_TOP_P
                    ),

                    max_tokens=(
                        CELL43_MAX_TOKENS
                    ),

                    extra_body={
                        "enable_thinking": False,
                    },
                )
            )

        except Exception as exc:

            attempt_history.append({
                "attempt": (
                    attempt_number
                ),

                "status": (
                    "API_ERROR"
                ),

                "error": (
                    f"{type(exc).__name__}: {exc}"
                ),
            })

            if (
                attempt_number
                < CELL43_MAX_ATTEMPTS
            ):

                wait_seconds = (
                    70.0
                    if cell37_is_rate_limit_exception(
                        exc
                    )
                    else (
                        CELL43_RETRY_SECONDS
                        * attempt_number
                    )
                )

                time.sleep(
                    wait_seconds
                )

                continue

            return {
                "candidate_id": (
                    request["candidate_id"]
                ),

                "execution_status": (
                    "API_FAILED"
                ),

                "attempt_history": (
                    attempt_history
                ),
            }

        elapsed_seconds = (
            time.time()
            - started_at
        )

        choice = response.choices[0]

        finish_reason = getattr(
            choice,
            "finish_reason",
            None,
        )

        raw_content = getattr(
            choice.message,
            "content",
            None,
        )

        usage = getattr(
            response,
            "usage",
            None,
        )

        input_tokens = int(
            getattr(
                usage,
                "prompt_tokens",
                0,
            )
            or 0
        )

        output_tokens = int(
            getattr(
                usage,
                "completion_tokens",
                0,
            )
            or 0
        )

        total_tokens = int(
            getattr(
                usage,
                "total_tokens",
                0,
            )
            or (
                input_tokens
                + output_tokens
            )
        )

        reasoning_tokens = 0

        completion_details = getattr(
            usage,
            "completion_tokens_details",
            None,
        )

        if completion_details is not None:

            reasoning_tokens = int(
                getattr(
                    completion_details,
                    "reasoning_tokens",
                    0,
                )
                or 0
            )

        clean_content = (
            cell37_clean_model_json(
                raw_content
            )
        )

        parsed_json = None
        parse_error = None

        try:

            parsed_json = json.loads(
                clean_content
            )

        except Exception as exc:

            parse_error = (
                f"{type(exc).__name__}: {exc}"
            )

        if parsed_json is None:

            structural_errors = [
                "JSON_PARSE_FAILED"
            ]

        else:

            structural_errors = (
                cell37_validate_output(
                    parsed_json
                )
            )

        structural_status = (
            "PASS"
            if not structural_errors
            else "FAIL"
        )

        semantic_verdict = (
            parsed_json.get(
                "verdict"
            )
            if (
                structural_status
                == "PASS"
            )
            else None
        )

        return {
            "candidate_id": (
                request["candidate_id"]
            ),

            "request_fingerprint": (
                request[
                    "request_fingerprint"
                ]
            ),

            "generated_qa_sha256": (
                request[
                    "generated_qa_sha256"
                ]
            ),

            "execution_status": (
                "COMPLETED"
            ),

            "finish_reason": (
                finish_reason
            ),

            "elapsed_seconds": round(
                elapsed_seconds,
                4,
            ),

            "usage": {
                "input_tokens": (
                    input_tokens
                ),

                "output_tokens": (
                    output_tokens
                ),

                "reasoning_tokens": (
                    reasoning_tokens
                ),

                "total_tokens": (
                    total_tokens
                ),
            },

            "estimated_cost_toman": (
                cell43_estimated_cost_toman(
                    input_tokens,
                    output_tokens,
                )
            ),

            "raw_content": (
                raw_content
            ),

            "parsed_json": (
                parsed_json
            ),

            "parse_error": (
                parse_error
            ),

            "structural_errors": (
                structural_errors
            ),

            "structural_status": (
                structural_status
            ),

            "semantic_verdict": (
                semantic_verdict
            ),

            "attempt_history": (
                attempt_history
            ),
        }

    raise RuntimeError(
        "Unexpected Cell 43 execution path."
    )


# =========================================================
# PREFLIGHT
# =========================================================

print(
    "=== GOLDEN V2 TARGETED V4 SEMANTIC VALIDATION PREFLIGHT ==="
)

print()

print(
    "Validation population:",
    len(cell43_requests),
)

print(
    "Candidates:",
    [
        request["candidate_id"]
        for request in cell43_requests
    ],
)

print(
    "Model:",
    semantic_review_model,
)

print(
    "Generation protocol:",
    GENERATION_V4_VERSION,
)

print(
    "Validation protocol:",
    CELL43_VALIDATION_PROTOCOL_VERSION,
)

print(
    "Max tokens:",
    CELL43_MAX_TOKENS,
)

print(
    "API allowed for missing exact checkpoints:",
    CELL43_ALLOW_API_FOR_MISSING,
)

print(
    "Previously accepted semantic results preserved:",
    10,
)

print(
    "Full Generation started:",
    False,
)

print()


# =========================================================
# RESOLVE EXACT TWO VALIDATIONS
# =========================================================

cell43_results = []
cell43_resolution_modes = []

cell43_api_calls_this_execution = 0
cell43_checkpoints_created_this_execution = 0


for index, request in enumerate(
    cell43_requests,
    start=1,
):

    (
        checkpoint,
        checkpoint_status,
        checkpoint_errors,
    ) = cell43_load_checkpoint(
        request
    )

    candidate_id = request[
        "candidate_id"
    ]

    if checkpoint_status == "VALID":

        result = checkpoint[
            "result"
        ]

        resolution_mode = (
            "EXACT_CHECKPOINT"
        )

    else:

        if checkpoint_status in {
            "INVALID",
            "UNREADABLE",
        }:

            raise AssertionError(
                "Existing Cell 43 checkpoint is invalid. "
                "Refusing automatic paid replacement for "
                f"{candidate_id}. "
                f"Status={checkpoint_status}; "
                f"errors={checkpoint_errors}"
            )

        if not CELL43_ALLOW_API_FOR_MISSING:

            raise AssertionError(
                "Exact Cell 43 checkpoint is missing and "
                "API execution is disabled for "
                f"{candidate_id}."
            )

        if (
            cell43_api_calls_this_execution
            > 0
        ):

            time.sleep(
                CELL43_REQUEST_SPACING_SECONDS
            )

        result = (
            cell43_execute_request(
                request
            )
        )

        cell43_api_calls_this_execution += 1

        checkpoint = (
            cell43_build_checkpoint(
                request,
                result,
            )
        )

        cell43_atomic_write_json(
            cell43_checkpoint_path(
                candidate_id
            ),
            checkpoint,
        )

        (
            verified_checkpoint,
            verified_status,
            verified_errors,
        ) = cell43_load_checkpoint(
            request
        )

        assert (
            verified_status
            == "VALID"
        ), (
            "New Cell 43 checkpoint verification failed "
            f"for {candidate_id}: "
            f"{verified_errors}"
        )

        result = (
            verified_checkpoint[
                "result"
            ]
        )

        resolution_mode = (
            "NEW_PAID_EXECUTION"
        )

        cell43_checkpoints_created_this_execution += 1

    cell43_results.append(
        result
    )

    cell43_resolution_modes.append({
        "candidate_id": (
            candidate_id
        ),

        "resolution_mode": (
            resolution_mode
        ),
    })

    print(
        f"[{index:02d}/02]",
        candidate_id,
        "|",
        resolution_mode,
        "| structural=",
        result.get(
            "structural_status",
            result.get(
                "execution_status"
            ),
        ),
        "| semantic=",
        result.get(
            "semantic_verdict"
        ),
        "| finish_reason=",
        result.get(
            "finish_reason"
        ),
        flush=True,
    )


# =========================================================
# FINAL DISK RESCAN
# =========================================================

cell43_valid_checkpoint_count = 0
cell43_missing_checkpoint_count = 0
cell43_invalid_checkpoint_count = 0


for request in cell43_requests:

    (
        checkpoint,
        status,
        errors,
    ) = cell43_load_checkpoint(
        request
    )

    if status == "VALID":

        cell43_valid_checkpoint_count += 1

    elif status == "MISSING":

        cell43_missing_checkpoint_count += 1

    else:

        cell43_invalid_checkpoint_count += 1


# =========================================================
# TARGETED RESULT COUNTS
# =========================================================

cell43_structural_pass_count = sum(
    result.get(
        "structural_status"
    )
    == "PASS"
    for result in cell43_results
)

cell43_api_failure_count = sum(
    result.get(
        "execution_status"
    )
    == "API_FAILED"
    for result in cell43_results
)

cell43_semantic_distribution = Counter(
    result.get(
        "semantic_verdict"
    )
    for result in cell43_results
    if result.get(
        "structural_status"
    )
    == "PASS"
)

cell43_dimension_failures = Counter()

for result in cell43_results:

    parsed = result.get(
        "parsed_json"
    )

    if not isinstance(
        parsed,
        dict,
    ):
        continue

    for field_name in CELL37_BOOLEAN_FIELDS:

        if parsed.get(
            field_name
        ) is False:

            cell43_dimension_failures[
                field_name
            ] += 1


cell43_known_costs = [
    result.get(
        "estimated_cost_toman"
    )
    for result in cell43_results
    if isinstance(
        result.get(
            "estimated_cost_toman"
        ),
        (int, float),
    )
]

cell43_total_estimated_cost_toman = (
    sum(cell43_known_costs)
    if cell43_known_costs
    else None
)


# =========================================================
# RECONSTRUCT FINAL 12 SEMANTIC RESULTS
# =========================================================

cell43_result_by_candidate_id = {
    result["candidate_id"]: result
    for result in cell43_results
}

assert len(
    cell43_result_by_candidate_id
) == 2


cell43_final_canary_validation_results = []

for historical_result in cell39_final_validation_results:

    candidate_id = (
        historical_result[
            "candidate_id"
        ]
    )

    if (
        candidate_id
        in cell43_result_by_candidate_id
    ):

        cell43_final_canary_validation_results.append(
            cell43_result_by_candidate_id[
                candidate_id
            ]
        )

    else:

        cell43_final_canary_validation_results.append(
            historical_result
        )


assert len(
    cell43_final_canary_validation_results
) == 12

assert len({
    result["candidate_id"]
    for result
    in cell43_final_canary_validation_results
}) == 12


cell43_final_structural_pass_count = sum(
    result.get(
        "structural_status"
    )
    == "PASS"
    for result
    in cell43_final_canary_validation_results
)

cell43_final_semantic_distribution = Counter(
    result.get(
        "semantic_verdict"
    )
    for result
    in cell43_final_canary_validation_results
    if result.get(
        "structural_status"
    )
    == "PASS"
)

cell43_final_semantic_pass_count = (
    cell43_final_semantic_distribution.get(
        "PASS",
        0,
    )
)

cell43_final_semantic_fail_count = (
    cell43_final_semantic_distribution.get(
        "FAIL",
        0,
    )
)


# =========================================================
# RUNNER HEALTH
# =========================================================

cell43_runner_checks = {
    "exactly_2_validation_targets": (
        len(cell43_requests)
        == 2
    ),

    "2_results_reconstructed": (
        len(cell43_results)
        == 2
    ),

    "2_valid_checkpoints": (
        cell43_valid_checkpoint_count
        == 2
    ),

    "zero_missing_checkpoints": (
        cell43_missing_checkpoint_count
        == 0
    ),

    "zero_invalid_checkpoints": (
        cell43_invalid_checkpoint_count
        == 0
    ),

    "zero_api_failures": (
        cell43_api_failure_count
        == 0
    ),

    "final_population_12": (
        len(
            cell43_final_canary_validation_results
        )
        == 12
    ),

    "full_generation_not_started": True,
}


cell43_runner_status = (
    "PASS"
    if all(
        cell43_runner_checks.values()
    )
    else "FAIL"
)


cell43_targeted_experimental_status = (
    "PASS"
    if (
        cell43_structural_pass_count
        == 2
        and
        cell43_semantic_distribution.get(
            "PASS",
            0,
        )
        == 2
    )
    else "FAIL"
)


cell43_final_canary_semantic_status = (
    "PASS"
    if (
        cell43_final_structural_pass_count
        == 12
        and
        cell43_final_semantic_pass_count
        == 12
    )
    else "FAIL"
)


if cell43_runner_status != "PASS":

    cell43_next_readiness = (
        "BLOCKED_CELL43_RUNNER_REPAIR_REQUIRED"
    )

elif (
    cell43_structural_pass_count
    < 2
):

    cell43_next_readiness = (
        "READY_FOR_CELL43_VALIDATOR_FAILURE_ANALYSIS"
    )

elif (
    cell43_semantic_distribution.get(
        "FAIL",
        0,
    )
    > 0
):

    cell43_next_readiness = (
        "READY_FOR_V4_SEMANTIC_FAILURE_ADJUDICATION"
    )

elif (
    cell43_final_canary_semantic_status
    == "PASS"
):

    cell43_next_readiness = (
        "READY_FOR_FULL_QA_GENERATION_V4"
    )

else:

    cell43_next_readiness = (
        "READY_FOR_FINAL_CANARY_STATE_ANALYSIS"
    )


# =========================================================
# OUTPUT
# =========================================================

print()

print(
    "=== GOLDEN V2 TARGETED V4 SEMANTIC VALIDATION RESULT ==="
)

print()

print(
    "Targeted results:",
    len(cell43_results),
)

print(
    "Valid checkpoints:",
    cell43_valid_checkpoint_count,
)

print(
    "Missing checkpoints:",
    cell43_missing_checkpoint_count,
)

print(
    "Invalid checkpoints:",
    cell43_invalid_checkpoint_count,
)

print(
    "Resolution modes:",
    dict(
        Counter(
            row["resolution_mode"]
            for row
            in cell43_resolution_modes
        )
    ),
)

print(
    "API calls this execution:",
    cell43_api_calls_this_execution,
)

print(
    "Checkpoints created this execution:",
    cell43_checkpoints_created_this_execution,
)

print(
    "Targeted Structural PASS:",
    cell43_structural_pass_count,
    "/ 2",
)

print(
    "Targeted Semantic distribution:",
    dict(
        cell43_semantic_distribution
    ),
)

print(
    "Targeted dimension failures:",
    dict(
        cell43_dimension_failures
    ),
)

print(
    "Estimated targeted validation cost (toman):",
    cell43_total_estimated_cost_toman,
)

print()


# =========================================================
# PRINT EXACT TARGETED JUDGMENTS
# =========================================================

for result in cell43_results:

    candidate_id = (
        result["candidate_id"]
    )

    generated = (
        cell43_generation_by_candidate_id[
            candidate_id
        ][
            "parsed_json"
        ]
    )

    print("=" * 78)

    print(
        "Candidate:",
        candidate_id,
    )

    print(
        "Question:",
        generated["question"],
    )

    print(
        "Expected Answer:",
        generated[
            "expected_answer"
        ],
    )

    print(
        "Semantic verdict:",
        result.get(
            "semantic_verdict"
        ),
    )

    print(
        "Validation:"
    )

    print(
        json.dumps(
            result.get(
                "parsed_json"
            ),
            ensure_ascii=False,
            indent=2,
        )
    )

    print()


# =========================================================
# FINAL CANARY STATE
# =========================================================

print("=" * 78)

print(
    "FINAL 12-CANDIDATE CANARY STATE"
)

print(
    "Structural PASS:",
    cell43_final_structural_pass_count,
    "/ 12",
)

print(
    "Semantic distribution:",
    dict(
        cell43_final_semantic_distribution
    ),
)

print(
    "Semantic PASS:",
    cell43_final_semantic_pass_count,
    "/ 12",
)

print(
    "Semantic FAIL:",
    cell43_final_semantic_fail_count,
    "/ 12",
)

print()


print(
    "Runner checks:"
)

for (
    check_name,
    check_value,
) in cell43_runner_checks.items():

    print(
        " -",
        check_name,
        ":",
        (
            "PASS"
            if check_value
            else "FAIL"
        ),
    )


print()

print(
    "Runner health:",
    cell43_runner_status,
)

print(
    "Targeted semantic result:",
    cell43_targeted_experimental_status,
)

print(
    "Final Canary semantic result:",
    cell43_final_canary_semantic_status,
)

print(
    "Previously accepted cases revalidated:",
    0,
)

print(
    "Full Generation started:",
    False,
)

print(
    "Next readiness:",
    cell43_next_readiness,
)

print(
    "CELL 43 CHECKPOINTED RUNNER:",
    cell43_runner_status,
)


assert cell43_runner_status == "PASS", (
    "Cell 43 checkpointed runner health failed."
)
````

<h2>مرحله ۴۴ — تولید کامل Question و Expected Answer با Generation v4</h2>
<p>
پس از موفقیت کامل Canary با ۱۲ از ۱۲ Semantic PASS، در این مرحله تولید نهایی Question و Expected Answer برای هر ۳۸۱۳ Golden Seed با قرارداد Generation v4 انجام می‌شود.
دو خروجی v4 معتبر که در Canary اصلاحی قبلاً تولید شده‌اند بدون فراخوانی مجدد API به Checkpointهای Full Generation منتقل می‌شوند.
تمام موارد دیگر دارای Request Fingerprint و Checkpoint مستقل هستند؛ بنابراین اجرای قطع‌شده قابل ادامه است و اجرای مجدد موارد تکمیل‌شده را دوباره به مدل ارسال نمی‌کند.
Checkpoint نامعتبر یا ناخوانا باعث توقف ایمن می‌شود و هرگز به‌صورت خودکار با یک درخواست پولی جایگزین نمی‌شود.
</p>

````python
# Code Cell 44 — Full QA Generation v4 Checkpointed Execution

from collections import Counter
from concurrent.futures import ThreadPoolExecutor, as_completed
from pathlib import Path
import hashlib
import json
import os
import re
import tempfile
import time


# =========================================================
# CONFIG
# =========================================================

CELL44_RUNNER_VERSION = (
    "golden_v2_full_qa_generation_v4_checkpoint_runner_v1"
)

CELL44_MAX_TOKENS = 1536
CELL44_TEMPERATURE = 0.7
CELL44_TOP_P = 0.8

CELL44_MAX_ATTEMPTS = 2
CELL44_RETRY_SECONDS = 2.0

# Conservative parallel execution.
CELL44_MAX_WORKERS = 4

# Intentional paid Full Generation.
# MUST become False before final Restart -> Run All.
CELL44_ALLOW_API_FOR_MISSING = False


# =========================================================
# REQUIRED STATE
# =========================================================

required_cell44_names = [
    "project_root",
    "semantic_review_client",
    "semantic_review_model",
    "cell25_candidate_skeletons",
    "cell41_build_generation_user_prompt_v4",
    "cell41_validate_generation_output_v4",
    "GOLDEN_V2_GENERATION_SYSTEM_PROMPT_V4",
    "GENERATION_V4_VERSION",
    "GENERATION_V4_PROMPT_VERSION",
    "GENERATION_V4_OUTPUT_SCHEMA_VERSION",
    "GENERATION_V4_PROTOCOL_SHA256",
    "cell42_repair_results",
]

missing_cell44_names = [
    name
    for name in required_cell44_names
    if name not in globals()
]

assert not missing_cell44_names, (
    "Missing required Cell 44 state: "
    + ", ".join(missing_cell44_names)
)

assert len(cell25_candidate_skeletons) == 3813

assert len({
    record["candidate_id"]
    for record in cell25_candidate_skeletons
}) == 3813

assert len(cell42_repair_results) == 2

assert all(
    result.get("structural_status") == "PASS"
    for result in cell42_repair_results
)

assert all(
    result.get("protocol_version")
    == GENERATION_V4_VERSION
    for result in cell42_repair_results
)

assert all(
    result.get("protocol_sha256")
    == GENERATION_V4_PROTOCOL_SHA256
    for result in cell42_repair_results
)


# =========================================================
# HASH UTILITIES
# =========================================================

def cell44_canonical_json(value):

    return json.dumps(
        value,
        ensure_ascii=False,
        sort_keys=True,
        separators=(",", ":"),
    )


def cell44_sha256_text(text):

    return hashlib.sha256(
        text.encode("utf-8")
    ).hexdigest()


# =========================================================
# RESPONSE CLEANING
# =========================================================

def cell44_clean_model_json(content):

    if not isinstance(content, str):
        return ""

    clean = content.strip()

    if clean.startswith("```"):

        clean = re.sub(
            r"^```(?:json)?\s*",
            "",
            clean,
            flags=re.IGNORECASE,
        )

        clean = re.sub(
            r"\s*```$",
            "",
            clean,
        )

    return clean.strip()


def cell44_is_rate_limit_exception(exc):

    error_text = str(exc).lower()
    error_type = type(exc).__name__.lower()

    return (
        "ratelimit" in error_type
        or "rate_limit" in error_type
        or "429" in error_text
        or "too many requests" in error_text
    )


# =========================================================
# EXACT REQUEST CONTRACT
# =========================================================

def cell44_build_request(candidate_id):

    user_prompt = (
        cell41_build_generation_user_prompt_v4(
            candidate_id
        )
    )

    contract = {
        "runner_version": (
            CELL44_RUNNER_VERSION
        ),

        "candidate_id": candidate_id,

        "protocol_version": (
            GENERATION_V4_VERSION
        ),

        "protocol_sha256": (
            GENERATION_V4_PROTOCOL_SHA256
        ),

        "prompt_version": (
            GENERATION_V4_PROMPT_VERSION
        ),

        "output_schema_version": (
            GENERATION_V4_OUTPUT_SCHEMA_VERSION
        ),

        "model": (
            semantic_review_model
        ),

        "system_prompt": (
            GOLDEN_V2_GENERATION_SYSTEM_PROMPT_V4
        ),

        "user_prompt": (
            user_prompt
        ),

        "temperature": (
            CELL44_TEMPERATURE
        ),

        "top_p": (
            CELL44_TOP_P
        ),

        "max_tokens": (
            CELL44_MAX_TOKENS
        ),

        "enable_thinking": False,
    }

    request_fingerprint = (
        cell44_sha256_text(
            cell44_canonical_json(
                contract
            )
        )
    )

    return {
        **contract,

        "request_fingerprint": (
            request_fingerprint
        ),
    }


cell44_candidate_ids = [
    record["candidate_id"]
    for record in cell25_candidate_skeletons
]

cell44_requests = [
    cell44_build_request(
        candidate_id
    )
    for candidate_id
    in cell44_candidate_ids
]

assert len(cell44_requests) == 3813

assert len({
    request["request_fingerprint"]
    for request in cell44_requests
}) == 3813


cell44_request_by_candidate_id = {
    request["candidate_id"]: request
    for request in cell44_requests
}

assert len(
    cell44_request_by_candidate_id
) == 3813


# =========================================================
# CHECKPOINT DIRECTORY
# =========================================================

cell44_checkpoint_dir = (
    Path(project_root)
    / "artifacts"
    / "phase2"
    / "golden_dataset_v2"
    / "work"
    / "full_qa_generation_v4"
)

cell44_checkpoint_dir.mkdir(
    parents=True,
    exist_ok=True,
)


def cell44_checkpoint_path(candidate_id):

    safe_id = re.sub(
        r"[^A-Za-z0-9_.-]+",
        "_",
        candidate_id,
    )

    return (
        cell44_checkpoint_dir
        / f"{safe_id}.json"
    )


# =========================================================
# ATOMIC WRITE
# =========================================================

def cell44_atomic_write_json(
    path,
    payload,
):

    path = Path(path)

    content = (
        json.dumps(
            payload,
            ensure_ascii=False,
            sort_keys=True,
            indent=2,
        )
        + "\n"
    )

    fd, temp_name = tempfile.mkstemp(
        prefix=path.name + ".",
        suffix=".tmp",
        dir=str(path.parent),
    )

    try:

        with os.fdopen(
            fd,
            "w",
            encoding="utf-8",
        ) as handle:

            handle.write(content)
            handle.flush()
            os.fsync(
                handle.fileno()
            )

        os.replace(
            temp_name,
            path,
        )

    except Exception:

        try:
            os.unlink(
                temp_name
            )
        except FileNotFoundError:
            pass

        raise


# =========================================================
# CHECKPOINT CONTRACT
# =========================================================

def cell44_build_checkpoint(
    request,
    result,
    provenance_mode,
):

    checkpoint = {
        "checkpoint_version": (
            CELL44_RUNNER_VERSION
        ),

        "candidate_id": (
            request["candidate_id"]
        ),

        "request_fingerprint": (
            request[
                "request_fingerprint"
            ]
        ),

        "protocol_version": (
            GENERATION_V4_VERSION
        ),

        "protocol_sha256": (
            GENERATION_V4_PROTOCOL_SHA256
        ),

        "prompt_version": (
            GENERATION_V4_PROMPT_VERSION
        ),

        "output_schema_version": (
            GENERATION_V4_OUTPUT_SCHEMA_VERSION
        ),

        "model": (
            semantic_review_model
        ),

        "temperature": (
            CELL44_TEMPERATURE
        ),

        "top_p": (
            CELL44_TOP_P
        ),

        "max_tokens": (
            CELL44_MAX_TOKENS
        ),

        "provenance_mode": (
            provenance_mode
        ),

        "result": result,
    }

    checkpoint[
        "checkpoint_payload_sha256"
    ] = cell44_sha256_text(
        cell44_canonical_json(
            checkpoint
        )
    )

    return checkpoint


def cell44_validate_checkpoint(
    checkpoint,
    request,
):

    errors = []

    if not isinstance(
        checkpoint,
        dict,
    ):
        return [
            "CHECKPOINT_NOT_OBJECT"
        ]

    stored_sha = checkpoint.get(
        "checkpoint_payload_sha256"
    )

    payload_without_sha = {
        key: value
        for key, value
        in checkpoint.items()
        if key
        != "checkpoint_payload_sha256"
    }

    recalculated_sha = (
        cell44_sha256_text(
            cell44_canonical_json(
                payload_without_sha
            )
        )
    )

    if stored_sha != recalculated_sha:

        errors.append(
            "CHECKPOINT_HASH_MISMATCH"
        )

    exact_expectations = {
        "checkpoint_version": (
            CELL44_RUNNER_VERSION
        ),

        "candidate_id": (
            request["candidate_id"]
        ),

        "request_fingerprint": (
            request[
                "request_fingerprint"
            ]
        ),

        "protocol_version": (
            GENERATION_V4_VERSION
        ),

        "protocol_sha256": (
            GENERATION_V4_PROTOCOL_SHA256
        ),

        "prompt_version": (
            GENERATION_V4_PROMPT_VERSION
        ),

        "output_schema_version": (
            GENERATION_V4_OUTPUT_SCHEMA_VERSION
        ),

        "model": (
            semantic_review_model
        ),

        "temperature": (
            CELL44_TEMPERATURE
        ),

        "top_p": (
            CELL44_TOP_P
        ),

        "max_tokens": (
            CELL44_MAX_TOKENS
        ),
    }

    for (
        field_name,
        expected_value,
    ) in exact_expectations.items():

        if (
            checkpoint.get(
                field_name
            )
            != expected_value
        ):

            errors.append(
                f"{field_name.upper()}_MISMATCH"
            )

    provenance_mode = checkpoint.get(
        "provenance_mode"
    )

    if provenance_mode not in {
        "FULL_V4_PAID_EXECUTION",
        "MIGRATED_TARGETED_V4_CANARY",
    }:

        errors.append(
            "INVALID_PROVENANCE_MODE"
        )

    result = checkpoint.get(
        "result"
    )

    if not isinstance(
        result,
        dict,
    ):

        errors.append(
            "CHECKPOINT_RESULT_INVALID"
        )

    elif (
        result.get("candidate_id")
        != request["candidate_id"]
    ):

        errors.append(
            "CHECKPOINT_RESULT_CANDIDATE_MISMATCH"
        )

    return sorted(
        set(errors)
    )


def cell44_load_checkpoint(request):

    path = cell44_checkpoint_path(
        request["candidate_id"]
    )

    if not path.exists():

        return (
            None,
            "MISSING",
            [],
        )

    try:

        with path.open(
            "r",
            encoding="utf-8",
        ) as handle:

            checkpoint = json.load(
                handle
            )

    except Exception as exc:

        return (
            None,
            "UNREADABLE",
            [
                f"{type(exc).__name__}: {exc}"
            ],
        )

    errors = (
        cell44_validate_checkpoint(
            checkpoint,
            request,
        )
    )

    if errors:

        return (
            checkpoint,
            "INVALID",
            errors,
        )

    return (
        checkpoint,
        "VALID",
        [],
    )


# =========================================================
# COST
# =========================================================

cell44_has_cost_rates = (
    "INPUT_TOMAN_PER_TOKEN" in globals()
    and
    "OUTPUT_TOMAN_PER_TOKEN" in globals()
)


def cell44_estimated_cost_toman(
    input_tokens,
    output_tokens,
):

    if not cell44_has_cost_rates:
        return None

    return (
        input_tokens
        * INPUT_TOMAN_PER_TOKEN
        +
        output_tokens
        * OUTPUT_TOMAN_PER_TOKEN
    )


# =========================================================
# API EXECUTION
# =========================================================

def cell44_execute_request(request):

    attempt_history = []
    api_attempts = 0

    for attempt_number in range(
        1,
        CELL44_MAX_ATTEMPTS + 1,
    ):

        started_at = time.time()
        api_attempts += 1

        try:

            response = (
                semantic_review_client
                .chat
                .completions
                .create(
                    model=(
                        semantic_review_model
                    ),

                    messages=[
                        {
                            "role": "system",
                            "content": (
                                GOLDEN_V2_GENERATION_SYSTEM_PROMPT_V4
                            ),
                        },
                        {
                            "role": "user",
                            "content": (
                                request[
                                    "user_prompt"
                                ]
                            ),
                        },
                    ],

                    temperature=(
                        CELL44_TEMPERATURE
                    ),

                    top_p=(
                        CELL44_TOP_P
                    ),

                    max_tokens=(
                        CELL44_MAX_TOKENS
                    ),

                    extra_body={
                        "enable_thinking": False,
                    },
                )
            )

        except Exception as exc:

            attempt_history.append({
                "attempt": (
                    attempt_number
                ),

                "status": (
                    "API_ERROR"
                ),

                "error": (
                    f"{type(exc).__name__}: {exc}"
                ),
            })

            if (
                attempt_number
                < CELL44_MAX_ATTEMPTS
            ):

                wait_seconds = (
                    70.0
                    if cell44_is_rate_limit_exception(
                        exc
                    )
                    else (
                        CELL44_RETRY_SECONDS
                        * attempt_number
                    )
                )

                time.sleep(
                    wait_seconds
                )

                continue

            return {
                "candidate_id": (
                    request["candidate_id"]
                ),

                "request_fingerprint": (
                    request[
                        "request_fingerprint"
                    ]
                ),

                "protocol_version": (
                    GENERATION_V4_VERSION
                ),

                "protocol_sha256": (
                    GENERATION_V4_PROTOCOL_SHA256
                ),

                "execution_status": (
                    "API_FAILED"
                ),

                "api_attempts": (
                    api_attempts
                ),

                "attempt_history": (
                    attempt_history
                ),

                "structural_status": (
                    "FAIL"
                ),

                "structural_errors": [
                    "API_FAILED"
                ],
            }

        elapsed_seconds = (
            time.time()
            - started_at
        )

        choice = response.choices[0]

        finish_reason = getattr(
            choice,
            "finish_reason",
            None,
        )

        raw_content = getattr(
            choice.message,
            "content",
            None,
        )

        usage = getattr(
            response,
            "usage",
            None,
        )

        input_tokens = int(
            getattr(
                usage,
                "prompt_tokens",
                0,
            )
            or 0
        )

        output_tokens = int(
            getattr(
                usage,
                "completion_tokens",
                0,
            )
            or 0
        )

        total_tokens = int(
            getattr(
                usage,
                "total_tokens",
                0,
            )
            or (
                input_tokens
                + output_tokens
            )
        )

        reasoning_tokens = 0

        completion_details = getattr(
            usage,
            "completion_tokens_details",
            None,
        )

        if completion_details is not None:

            reasoning_tokens = int(
                getattr(
                    completion_details,
                    "reasoning_tokens",
                    0,
                )
                or 0
            )

        clean_content = (
            cell44_clean_model_json(
                raw_content
            )
        )

        parsed_json = None
        parse_error = None

        try:

            parsed_json = json.loads(
                clean_content
            )

        except Exception as exc:

            parse_error = (
                f"{type(exc).__name__}: {exc}"
            )

        if parsed_json is None:

            structural_errors = [
                "JSON_PARSE_FAILED"
            ]

        else:

            structural_errors = (
                cell41_validate_generation_output_v4(
                    parsed_json
                )
            )

        structural_status = (
            "PASS"
            if not structural_errors
            else "FAIL"
        )

        return {
            "candidate_id": (
                request["candidate_id"]
            ),

            "request_fingerprint": (
                request[
                    "request_fingerprint"
                ]
            ),

            "protocol_version": (
                GENERATION_V4_VERSION
            ),

            "protocol_sha256": (
                GENERATION_V4_PROTOCOL_SHA256
            ),

            "execution_status": (
                "COMPLETED"
            ),

            "api_attempts": (
                api_attempts
            ),

            "finish_reason": (
                finish_reason
            ),

            "elapsed_seconds": round(
                elapsed_seconds,
                4,
            ),

            "usage": {
                "input_tokens": (
                    input_tokens
                ),

                "output_tokens": (
                    output_tokens
                ),

                "reasoning_tokens": (
                    reasoning_tokens
                ),

                "total_tokens": (
                    total_tokens
                ),
            },

            "estimated_cost_toman": (
                cell44_estimated_cost_toman(
                    input_tokens,
                    output_tokens,
                )
            ),

            "raw_content": (
                raw_content
            ),

            "parsed_json": (
                parsed_json
            ),

            "parse_error": (
                parse_error
            ),

            "structural_errors": (
                structural_errors
            ),

            "structural_status": (
                structural_status
            ),

            "attempt_history": (
                attempt_history
            ),
        }

    raise RuntimeError(
        "Unexpected Cell 44 execution path."
    )


# =========================================================
# PREFLIGHT SCAN
# IMPORTANT:
# Detect invalid checkpoints BEFORE any paid request.
# =========================================================

cell44_existing_valid = {}
cell44_missing_requests = []
cell44_invalid_existing = []


for request in cell44_requests:

    (
        checkpoint,
        status,
        errors,
    ) = cell44_load_checkpoint(
        request
    )

    candidate_id = request[
        "candidate_id"
    ]

    if status == "VALID":

        cell44_existing_valid[
            candidate_id
        ] = checkpoint

    elif status == "MISSING":

        cell44_missing_requests.append(
            request
        )

    else:

        cell44_invalid_existing.append({
            "candidate_id": (
                candidate_id
            ),

            "status": (
                status
            ),

            "errors": (
                errors
            ),
        })


assert not cell44_invalid_existing, (
    "Invalid/unreadable Cell 44 checkpoints detected. "
    "Refusing paid execution. "
    f"Examples: {cell44_invalid_existing[:5]}"
)


# =========================================================
# MIGRATE THE TWO EXISTING TARGETED v4 GENERATIONS
# =========================================================

cell44_targeted_v4_by_candidate_id = {
    result["candidate_id"]: result
    for result in cell42_repair_results
}

assert len(
    cell44_targeted_v4_by_candidate_id
) == 2


cell44_migrated_this_execution = 0


for candidate_id, historical_result in (
    cell44_targeted_v4_by_candidate_id.items()
):

    request = (
        cell44_request_by_candidate_id[
            candidate_id
        ]
    )

    (
        existing_checkpoint,
        existing_status,
        existing_errors,
    ) = cell44_load_checkpoint(
        request
    )

    if existing_status == "VALID":
        continue

    assert existing_status == "MISSING", (
        "Refusing migration over invalid checkpoint for "
        f"{candidate_id}: "
        f"{existing_status} {existing_errors}"
    )

    assert (
        historical_result.get(
            "protocol_version"
        )
        == GENERATION_V4_VERSION
    )

    assert (
        historical_result.get(
            "protocol_sha256"
        )
        == GENERATION_V4_PROTOCOL_SHA256
    )

    assert (
        historical_result.get(
            "structural_status"
        )
        == "PASS"
    )

    migrated_result = dict(
        historical_result
    )

    migrated_result[
        "request_fingerprint"
    ] = request[
        "request_fingerprint"
    ]

    checkpoint = (
        cell44_build_checkpoint(
            request,
            migrated_result,
            "MIGRATED_TARGETED_V4_CANARY",
        )
    )

    cell44_atomic_write_json(
        cell44_checkpoint_path(
            candidate_id
        ),
        checkpoint,
    )

    (
        verified_checkpoint,
        verified_status,
        verified_errors,
    ) = cell44_load_checkpoint(
        request
    )

    assert verified_status == "VALID", (
        "Cell 44 v4 Canary migration verification failed "
        f"for {candidate_id}: "
        f"{verified_errors}"
    )

    cell44_migrated_this_execution += 1


# =========================================================
# RESCAN AFTER MIGRATION
# =========================================================

cell44_existing_valid = {}
cell44_requests_for_api = []
cell44_invalid_after_migration = []


for request in cell44_requests:

    (
        checkpoint,
        status,
        errors,
    ) = cell44_load_checkpoint(
        request
    )

    candidate_id = request[
        "candidate_id"
    ]

    if status == "VALID":

        cell44_existing_valid[
            candidate_id
        ] = checkpoint

    elif status == "MISSING":

        cell44_requests_for_api.append(
            request
        )

    else:

        cell44_invalid_after_migration.append({
            "candidate_id": (
                candidate_id
            ),

            "status": (
                status
            ),

            "errors": (
                errors
            ),
        })


assert not cell44_invalid_after_migration, (
    "Invalid checkpoints detected after migration."
)


# =========================================================
# PREFLIGHT OUTPUT
# =========================================================

print(
    "=== GOLDEN V2 FULL QA GENERATION V4 PREFLIGHT ==="
)

print()

print(
    "Target population:",
    len(cell44_requests),
)

print(
    "Exact valid checkpoints before API:",
    len(cell44_existing_valid),
)

print(
    "Targeted v4 Canary checkpoints migrated this execution:",
    cell44_migrated_this_execution,
)

print(
    "Remaining candidates requiring API:",
    len(cell44_requests_for_api),
)

print(
    "Model:",
    semantic_review_model,
)

print(
    "Protocol:",
    GENERATION_V4_VERSION,
)

print(
    "Protocol SHA:",
    GENERATION_V4_PROTOCOL_SHA256,
)

print(
    "Max tokens:",
    CELL44_MAX_TOKENS,
)

print(
    "Parallel workers:",
    CELL44_MAX_WORKERS,
)

print(
    "API allowed for missing exact checkpoints:",
    CELL44_ALLOW_API_FOR_MISSING,
)

print()


if (
    cell44_requests_for_api
    and not CELL44_ALLOW_API_FOR_MISSING
):

    raise AssertionError(
        "Cell 44 has missing Full Generation checkpoints "
        "but API execution is disabled."
    )


# =========================================================
# FULL GENERATION
# =========================================================

cell44_api_candidates_this_execution = 0
cell44_api_attempts_this_execution = 0
cell44_checkpoints_created_this_execution = (
    cell44_migrated_this_execution
)


def cell44_execute_and_checkpoint(
    request,
):

    candidate_id = request[
        "candidate_id"
    ]

    result = (
        cell44_execute_request(
            request
        )
    )

    checkpoint = (
        cell44_build_checkpoint(
            request,
            result,
            "FULL_V4_PAID_EXECUTION",
        )
    )

    cell44_atomic_write_json(
        cell44_checkpoint_path(
            candidate_id
        ),
        checkpoint,
    )

    (
        verified_checkpoint,
        verified_status,
        verified_errors,
    ) = cell44_load_checkpoint(
        request
    )

    if verified_status != "VALID":

        raise AssertionError(
            "Cell 44 checkpoint verification failed "
            f"for {candidate_id}: "
            f"{verified_errors}"
        )

    return (
        candidate_id,
        verified_checkpoint[
            "result"
        ],
    )


if cell44_requests_for_api:

    completed_count = 0

    with ThreadPoolExecutor(
        max_workers=CELL44_MAX_WORKERS
    ) as executor:

        future_to_request = {
            executor.submit(
                cell44_execute_and_checkpoint,
                request,
            ): request
            for request
            in cell44_requests_for_api
        }

        for future in as_completed(
            future_to_request
        ):

            request = (
                future_to_request[
                    future
                ]
            )

            candidate_id = request[
                "candidate_id"
            ]

            try:

                (
                    resolved_candidate_id,
                    result,
                ) = future.result()

            except Exception as exc:

                raise RuntimeError(
                    "Cell 44 worker/checkpoint failure for "
                    f"{candidate_id}: "
                    f"{type(exc).__name__}: {exc}"
                ) from exc

            assert (
                resolved_candidate_id
                == candidate_id
            )

            completed_count += 1

            cell44_api_candidates_this_execution += 1

            cell44_api_attempts_this_execution += int(
                result.get(
                    "api_attempts",
                    0,
                )
                or 0
            )

            cell44_checkpoints_created_this_execution += 1

            print(
                f"[{completed_count:04d}/"
                f"{len(cell44_requests_for_api):04d}]",
                candidate_id,
                "| structural=",
                result.get(
                    "structural_status"
                ),
                "| finish=",
                result.get(
                    "finish_reason"
                ),
                flush=True,
            )


# =========================================================
# FINAL DISK RESCAN
# =========================================================

cell44_full_generation_results = []

cell44_valid_checkpoint_count = 0
cell44_missing_checkpoint_count = 0
cell44_invalid_checkpoint_count = 0

cell44_provenance_distribution = Counter()


for request in cell44_requests:

    (
        checkpoint,
        status,
        errors,
    ) = cell44_load_checkpoint(
        request
    )

    if status == "VALID":

        cell44_valid_checkpoint_count += 1

        cell44_provenance_distribution[
            checkpoint[
                "provenance_mode"
            ]
        ] += 1

        cell44_full_generation_results.append(
            checkpoint[
                "result"
            ]
        )

    elif status == "MISSING":

        cell44_missing_checkpoint_count += 1

    else:

        cell44_invalid_checkpoint_count += 1


# =========================================================
# FINAL COUNTS
# =========================================================

cell44_structural_distribution = Counter(
    result.get(
        "structural_status"
    )
    for result
    in cell44_full_generation_results
)

cell44_finish_reason_distribution = Counter(
    result.get(
        "finish_reason"
    )
    for result
    in cell44_full_generation_results
)

cell44_execution_status_distribution = Counter(
    result.get(
        "execution_status"
    )
    for result
    in cell44_full_generation_results
)


cell44_structural_pass_count = (
    cell44_structural_distribution.get(
        "PASS",
        0,
    )
)

cell44_structural_fail_count = (
    cell44_structural_distribution.get(
        "FAIL",
        0,
    )
)


cell44_api_failure_count = (
    cell44_execution_status_distribution.get(
        "API_FAILED",
        0,
    )
)


cell44_known_costs = [
    result.get(
        "estimated_cost_toman"
    )
    for result
    in cell44_full_generation_results
    if isinstance(
        result.get(
            "estimated_cost_toman"
        ),
        (int, float),
    )
]

cell44_total_estimated_cost_toman = (
    sum(
        cell44_known_costs
    )
    if cell44_known_costs
    else None
)


# =========================================================
# IDENTITY CHECK
# =========================================================

cell44_result_candidate_ids = [
    result["candidate_id"]
    for result
    in cell44_full_generation_results
]

cell44_identity_exact = (
    len(
        cell44_result_candidate_ids
    )
    == 3813
    and
    len(
        set(
            cell44_result_candidate_ids
        )
    )
    == 3813
    and
    set(
        cell44_result_candidate_ids
    )
    == set(
        cell44_candidate_ids
    )
)


# =========================================================
# RUNNER HEALTH
# =========================================================

cell44_runner_checks = {
    "3813_requests_defined": (
        len(cell44_requests)
        == 3813
    ),

    "3813_results_reconstructed": (
        len(
            cell44_full_generation_results
        )
        == 3813
    ),

    "3813_valid_checkpoints": (
        cell44_valid_checkpoint_count
        == 3813
    ),

    "zero_missing_checkpoints": (
        cell44_missing_checkpoint_count
        == 0
    ),

    "zero_invalid_checkpoints": (
        cell44_invalid_checkpoint_count
        == 0
    ),

    "exact_candidate_identity": (
        cell44_identity_exact
    ),

    "checkpoint_count_matches_population": (
        sum(
            cell44_provenance_distribution.values()
        )
        == 3813
    ),
}


cell44_runner_status = (
    "PASS"
    if all(
        cell44_runner_checks.values()
    )
    else "FAIL"
)


# Experimental generation quality is separate
# from checkpoint/runner integrity.
cell44_generation_structural_status = (
    "PASS"
    if (
        cell44_structural_pass_count
        == 3813
        and
        cell44_api_failure_count
        == 0
    )
    else "FAIL"
)


if cell44_runner_status != "PASS":

    cell44_next_readiness = (
        "BLOCKED_CELL44_RUNNER_REPAIR_REQUIRED"
    )

elif cell44_api_failure_count > 0:

    cell44_next_readiness = (
        "READY_FOR_FULL_GENERATION_API_FAILURE_RECOVERY"
    )

elif cell44_structural_fail_count > 0:

    cell44_next_readiness = (
        "READY_FOR_FULL_GENERATION_STRUCTURAL_FAILURE_ANALYSIS"
    )

else:

    cell44_next_readiness = (
        "READY_FOR_FULL_QA_SEMANTIC_VALIDATION_V4"
    )


# =========================================================
# OUTPUT
# =========================================================

print()

print(
    "=== GOLDEN V2 FULL QA GENERATION V4 RESULT ==="
)

print()

print(
    "Target population:",
    len(cell44_requests),
)

print(
    "Results reconstructed:",
    len(
        cell44_full_generation_results
    ),
)

print(
    "Valid checkpoints:",
    cell44_valid_checkpoint_count,
)

print(
    "Missing checkpoints:",
    cell44_missing_checkpoint_count,
)

print(
    "Invalid checkpoints:",
    cell44_invalid_checkpoint_count,
)

print(
    "Checkpoint provenance:",
    dict(
        cell44_provenance_distribution
    ),
)

print()

print(
    "API candidates this execution:",
    cell44_api_candidates_this_execution,
)

print(
    "API attempts this execution:",
    cell44_api_attempts_this_execution,
)

print(
    "Checkpoints created/migrated this execution:",
    cell44_checkpoints_created_this_execution,
)

print()

print(
    "Structural distribution:",
    dict(
        cell44_structural_distribution
    ),
)

print(
    "Structural PASS:",
    cell44_structural_pass_count,
    "/ 3813",
)

print(
    "Structural FAIL:",
    cell44_structural_fail_count,
    "/ 3813",
)

print(
    "API failures:",
    cell44_api_failure_count,
)

print(
    "Finish reason distribution:",
    dict(
        cell44_finish_reason_distribution
    ),
)

print(
    "Estimated total generation cost represented "
    "by checkpoints (toman):",
    cell44_total_estimated_cost_toman,
)

print()

print(
    "Runner checks:"
)

for (
    check_name,
    check_value,
) in cell44_runner_checks.items():

    print(
        " -",
        check_name,
        ":",
        (
            "PASS"
            if check_value
            else "FAIL"
        ),
    )

print()

print(
    "Runner health:",
    cell44_runner_status,
)

print(
    "Full Generation structural result:",
    cell44_generation_structural_status,
)

print(
    "Next readiness:",
    cell44_next_readiness,
)

print(
    "CELL 44 CHECKPOINTED RUNNER:",
    cell44_runner_status,
)


assert cell44_runner_status == "PASS", (
    "Cell 44 checkpointed runner health failed."
)
````

<h2>مرحله ۴۵ — تفکیک و آماده‌سازی بازیابی شکست‌های Full Generation</h2>
<p>
اجرای کامل Generation v4 برای هر ۳۸۱۳ کاندید Checkpoint معتبر ساخته است، اما ۱۸۲ خروجی از نظر ساختاری نامعتبر مانده‌اند.
در این مرحله بدون هیچ فراخوانی API، شکست‌ها به دو گروه اصلی تفکیک می‌شوند:
موارد متوقف‌شده به علت محدودیت طول خروجی و موارد دارای خطای API.
هدف، ساخت دقیق صف بازیابی فقط برای همین موارد و جلوگیری از تکرار ۳۶۳۱ تولید موفق است.
</p>
</div>

````python
# Code Cell 45 — Full Generation Failure Recovery Preparation

from collections import Counter


# =========================================================
# SAFETY
# =========================================================

CELL45_API_CALLS = 0
CELL45_FILE_WRITES = 0


# =========================================================
# DISCOVER THE EXACT CELL 44 RESULT COLLECTION
#
# We deliberately identify it by the observed Cell 44
# invariants instead of guessing its variable name.
# =========================================================

cell45_result_candidates = []


for variable_name, value in list(globals().items()):

    if not isinstance(value, list):
        continue

    if len(value) != 3813:
        continue

    if not all(
        isinstance(row, dict)
        for row in value
    ):
        continue

    if not all(
        row.get("candidate_id")
        for row in value
    ):
        continue

    structural_values = Counter(
        row.get("structural_status")
        for row in value
    )

    if (
        structural_values.get("PASS", 0) == 3631
        and
        structural_values.get("FAIL", 0) == 182
    ):

        cell45_result_candidates.append(
            (
                variable_name,
                value,
            )
        )


assert cell45_result_candidates, (
    "Could not locate the exact Cell 44 "
    "3813-result collection."
)


# Prefer an explicitly Cell-44-named object when available.
cell45_cell44_named_candidates = [
    item
    for item in cell45_result_candidates
    if "cell44" in item[0].lower()
]


if cell45_cell44_named_candidates:

    cell45_source_variable_name, cell45_cell44_results = (
        cell45_cell44_named_candidates[0]
    )

else:

    # Multiple aliases of the same result collection are harmless
    # only when candidate/status signatures are identical.
    cell45_signatures = []

    for variable_name, value in cell45_result_candidates:

        signature = tuple(
            (
                row.get("candidate_id"),
                row.get("structural_status"),
                row.get("finish_reason"),
            )
            for row in value
        )

        cell45_signatures.append(
            (
                variable_name,
                signature,
                value,
            )
        )

    unique_signatures = {
        item[1]
        for item in cell45_signatures
    }

    assert len(unique_signatures) == 1, (
        "More than one non-equivalent 3813-result collection "
        "matches Cell 44 invariants."
    )

    (
        cell45_source_variable_name,
        _,
        cell45_cell44_results,
    ) = cell45_signatures[0]


# =========================================================
# EXACT FAILURE POPULATION
# =========================================================

cell45_failed_results = [
    row
    for row in cell45_cell44_results
    if row.get("structural_status") == "FAIL"
]


cell45_pass_results = [
    row
    for row in cell45_cell44_results
    if row.get("structural_status") == "PASS"
]


assert len(cell45_cell44_results) == 3813
assert len(cell45_pass_results) == 3631
assert len(cell45_failed_results) == 182


# =========================================================
# FAILURE CLASSIFICATION
# =========================================================

cell45_length_failures = [
    row
    for row in cell45_failed_results
    if row.get("finish_reason") == "length"
]


cell45_non_length_failures = [
    row
    for row in cell45_failed_results
    if row.get("finish_reason") != "length"
]


assert len(cell45_length_failures) == 175
assert len(cell45_non_length_failures) == 7


cell45_length_recovery_candidate_ids = [
    row["candidate_id"]
    for row in cell45_length_failures
]


cell45_api_recovery_candidate_ids = [
    row["candidate_id"]
    for row in cell45_non_length_failures
]


cell45_all_recovery_candidate_ids = (
    cell45_length_recovery_candidate_ids
    +
    cell45_api_recovery_candidate_ids
)


assert len(cell45_all_recovery_candidate_ids) == 182

assert (
    len(
        set(
            cell45_all_recovery_candidate_ids
        )
    )
    == 182
)


# =========================================================
# DIAGNOSTICS
# =========================================================

cell45_finish_reason_distribution = Counter(
    row.get("finish_reason")
    for row in cell45_failed_results
)


cell45_execution_status_distribution = Counter(
    row.get("execution_status")
    for row in cell45_failed_results
)


cell45_non_length_execution_status_distribution = Counter(
    row.get("execution_status")
    for row in cell45_non_length_failures
)


cell45_error_type_distribution = Counter()


for row in cell45_non_length_failures:

    error_value = row.get("error")

    if error_value:

        error_type = (
            str(error_value)
            .split(":", 1)[0]
            .strip()
        )

    else:

        error_type = "NO_ERROR_TEXT"

    cell45_error_type_distribution[
        error_type
    ] += 1


# =========================================================
# RECOVERY POLICY
# =========================================================
#
# 175 length failures:
#   regenerate only these candidates with a larger output cap.
#
# 7 API failures:
#   retry only these candidates independently.
#
# 3631 successful generations:
#   never regenerate.
#
# No semantic validation begins until all 3813 generated
# outputs are structurally valid.
# =========================================================

cell45_recovery_policy = {
    "successful_generation_population": 3631,

    "recovery_population": 182,

    "length_recovery_population": 175,

    "api_failure_recovery_population": 7,

    "preserve_successful_generations": True,

    "regenerate_successful_candidates": False,

    "semantic_validation_started": False,

    "length_recovery_requires_larger_output_cap": True,

    "api_failure_recovery_is_targeted_only": True,
}


# =========================================================
# VALIDATION
# =========================================================

cell45_checks = {
    "3813_cell44_results_found": (
        len(cell45_cell44_results) == 3813
    ),

    "3631_successful_generations_preserved": (
        len(cell45_pass_results) == 3631
    ),

    "182_failures_identified": (
        len(cell45_failed_results) == 182
    ),

    "175_length_failures_identified": (
        len(cell45_length_failures) == 175
    ),

    "7_non_length_failures_identified": (
        len(cell45_non_length_failures) == 7
    ),

    "182_unique_recovery_candidates": (
        len(
            set(
                cell45_all_recovery_candidate_ids
            )
        )
        == 182
    ),

    "zero_api_calls": (
        CELL45_API_CALLS == 0
    ),

    "zero_file_writes": (
        CELL45_FILE_WRITES == 0
    ),

    "semantic_validation_not_started": True,
}


cell45_status = (
    "PASS"
    if all(
        cell45_checks.values()
    )
    else "FAIL"
)


cell45_next_readiness = (
    "READY_FOR_TARGETED_FULL_GENERATION_RECOVERY"
    if cell45_status == "PASS"
    else "BLOCKED_CELL45_REPAIR_REQUIRED"
)


# =========================================================
# OUTPUT
# =========================================================

print()
print(
    "=== GOLDEN V2 FULL GENERATION FAILURE "
    "RECOVERY PREPARATION ==="
)
print()

print(
    "Cell 44 source variable:",
    cell45_source_variable_name,
)

print(
    "Full population:",
    len(cell45_cell44_results),
)

print(
    "Structural PASS preserved:",
    len(cell45_pass_results),
)

print(
    "Recovery population:",
    len(cell45_failed_results),
)

print()

print(
    "Failure finish-reason distribution:",
    dict(cell45_finish_reason_distribution),
)

print(
    "Failure execution-status distribution:",
    dict(cell45_execution_status_distribution),
)

print()

print(
    "Length recovery candidates:",
    len(cell45_length_recovery_candidate_ids),
)

print(
    "API/non-length recovery candidates:",
    len(cell45_api_recovery_candidate_ids),
)

print(
    "API/non-length execution statuses:",
    dict(
        cell45_non_length_execution_status_distribution
    ),
)

print(
    "API/non-length error types:",
    dict(cell45_error_type_distribution),
)

print()

print("Validation checks:")

for check_name, check_value in cell45_checks.items():

    print(
        " -",
        check_name,
        ":",
        "PASS" if check_value else "FAIL",
    )

print()

print(
    "API calls:",
    CELL45_API_CALLS,
)

print(
    "File writes:",
    CELL45_FILE_WRITES,
)

print(
    "CELL 45 STATUS:",
    cell45_status,
)

print(
    "Next readiness:",
    cell45_next_readiness,
)


assert cell45_status == "PASS", (
    "Cell 45 recovery preparation failed."
)
````

<h2>مرحله ۴۶ — بازیابی هدفمند شکست‌های Full Generation v4</h2>
<p>
در این مرحله فقط ۱۸۲ خروجی ناموفق Full Generation بازیابی می‌شوند و ۳۶۳۱ خروجی موفق قبلی بدون تغییر حفظ خواهند شد.
برای ۱۷۵ موردی که به علت محدودیت طول متوقف شده‌اند و ۷ مورد دارای خطای API، درخواست‌های Recovery مستقل با Checkpoint و Fingerprint اختصاصی ساخته می‌شود.
سقف خروجی Recovery به ۳۰۷۲ توکن افزایش می‌یابد و پس از پایان، نتایج بازیابی‌شده با ۳۶۳۱ خروجی سالم قبلی ترکیب می‌شوند تا جمعیت نهایی ۳۸۱۳ مورد دوباره بررسی شود.
</p>
</div>

````python
# Code Cell 46 — Targeted Full Generation v4 Recovery

from collections import Counter
from concurrent.futures import ThreadPoolExecutor, as_completed
from pathlib import Path
import hashlib
import json
import os
import re
import tempfile
import time


# =========================================================
# CONFIG
# =========================================================

CELL46_RUNNER_VERSION = (
    "golden_v2_full_generation_v4_targeted_recovery_v1"
)

CELL46_MAX_TOKENS = 3072
CELL46_TEMPERATURE = 0.7
CELL46_TOP_P = 0.8

CELL46_MAX_ATTEMPTS = 2
CELL46_RETRY_SECONDS = 2.0

# Conservative recovery concurrency.
CELL46_MAX_WORKERS = 2

# Intentional paid recovery of ONLY the 182 failed candidates.
# MUST become False before final Restart -> Run All.
CELL46_ALLOW_API_FOR_MISSING = False


# =========================================================
# REQUIRED STATE
# =========================================================

required_cell46_names = [
    "project_root",
    "semantic_review_client",
    "semantic_review_model",
    "cell44_full_generation_results",
    "cell44_request_by_candidate_id",
    "cell45_pass_results",
    "cell45_failed_results",
    "cell45_all_recovery_candidate_ids",
    "cell45_length_recovery_candidate_ids",
    "cell45_api_recovery_candidate_ids",
    "cell41_validate_generation_output_v4",
    "GOLDEN_V2_GENERATION_SYSTEM_PROMPT_V4",
    "GENERATION_V4_VERSION",
    "GENERATION_V4_PROMPT_VERSION",
    "GENERATION_V4_OUTPUT_SCHEMA_VERSION",
    "GENERATION_V4_PROTOCOL_SHA256",
]

missing_cell46_names = [
    name
    for name in required_cell46_names
    if name not in globals()
]

assert not missing_cell46_names, (
    "Missing required Cell 46 state: "
    + ", ".join(missing_cell46_names)
)

assert len(cell44_full_generation_results) == 3813
assert len(cell45_pass_results) == 3631
assert len(cell45_failed_results) == 182

assert len(
    cell45_length_recovery_candidate_ids
) == 175

assert len(
    cell45_api_recovery_candidate_ids
) == 7

assert len(
    cell45_all_recovery_candidate_ids
) == 182

assert len(
    set(
        cell45_all_recovery_candidate_ids
    )
) == 182


# =========================================================
# HASH UTILITIES
# =========================================================

def cell46_canonical_json(value):

    return json.dumps(
        value,
        ensure_ascii=False,
        sort_keys=True,
        separators=(",", ":"),
    )


def cell46_sha256_text(text):

    return hashlib.sha256(
        text.encode("utf-8")
    ).hexdigest()


# =========================================================
# MODEL OUTPUT CLEANING
# =========================================================

def cell46_clean_model_json(content):

    if not isinstance(content, str):
        return ""

    clean = content.strip()

    if clean.startswith("```"):

        clean = re.sub(
            r"^```(?:json)?\s*",
            "",
            clean,
            flags=re.IGNORECASE,
        )

        clean = re.sub(
            r"\s*```$",
            "",
            clean,
        )

    return clean.strip()


def cell46_is_rate_limit_exception(exc):

    error_text = str(exc).lower()
    error_type = type(exc).__name__.lower()

    return (
        "ratelimit" in error_type
        or "rate_limit" in error_type
        or "429" in error_text
        or "too many requests" in error_text
    )


# =========================================================
# ORIGINAL FAILURE LOOKUP
# =========================================================

cell46_original_failure_by_candidate_id = {
    row["candidate_id"]: row
    for row in cell45_failed_results
}

assert len(
    cell46_original_failure_by_candidate_id
) == 182


# =========================================================
# EXACT RECOVERY REQUESTS
# =========================================================

def cell46_build_request(candidate_id):

    original_request = (
        cell44_request_by_candidate_id[
            candidate_id
        ]
    )

    original_failure = (
        cell46_original_failure_by_candidate_id[
            candidate_id
        ]
    )

    if candidate_id in set(
        cell45_length_recovery_candidate_ids
    ):
        recovery_reason = "LENGTH"
    else:
        recovery_reason = "API_FAILED"

    contract = {
        "runner_version": (
            CELL46_RUNNER_VERSION
        ),

        "candidate_id": candidate_id,

        "recovery_reason": (
            recovery_reason
        ),

        "original_request_fingerprint": (
            original_request[
                "request_fingerprint"
            ]
        ),

        "original_finish_reason": (
            original_failure.get(
                "finish_reason"
            )
        ),

        "original_execution_status": (
            original_failure.get(
                "execution_status"
            )
        ),

        "protocol_version": (
            GENERATION_V4_VERSION
        ),

        "protocol_sha256": (
            GENERATION_V4_PROTOCOL_SHA256
        ),

        "prompt_version": (
            GENERATION_V4_PROMPT_VERSION
        ),

        "output_schema_version": (
            GENERATION_V4_OUTPUT_SCHEMA_VERSION
        ),

        "model": (
            semantic_review_model
        ),

        "system_prompt": (
            GOLDEN_V2_GENERATION_SYSTEM_PROMPT_V4
        ),

        "user_prompt": (
            original_request[
                "user_prompt"
            ]
        ),

        "temperature": (
            CELL46_TEMPERATURE
        ),

        "top_p": (
            CELL46_TOP_P
        ),

        "max_tokens": (
            CELL46_MAX_TOKENS
        ),

        "enable_thinking": False,
    }

    request_fingerprint = (
        cell46_sha256_text(
            cell46_canonical_json(
                contract
            )
        )
    )

    return {
        **contract,

        "request_fingerprint": (
            request_fingerprint
        ),
    }


cell46_requests = [
    cell46_build_request(
        candidate_id
    )
    for candidate_id
    in cell45_all_recovery_candidate_ids
]

assert len(cell46_requests) == 182

assert len({
    request["candidate_id"]
    for request in cell46_requests
}) == 182

assert len({
    request["request_fingerprint"]
    for request in cell46_requests
}) == 182


cell46_request_by_candidate_id = {
    request["candidate_id"]: request
    for request in cell46_requests
}


# =========================================================
# CHECKPOINT DIRECTORY
# =========================================================

cell46_checkpoint_dir = (
    Path(project_root)
    / "artifacts"
    / "phase2"
    / "golden_dataset_v2"
    / "work"
    / "full_qa_generation_v4_recovery_3072"
)

cell46_checkpoint_dir.mkdir(
    parents=True,
    exist_ok=True,
)


def cell46_checkpoint_path(candidate_id):

    safe_id = re.sub(
        r"[^A-Za-z0-9_.-]+",
        "_",
        candidate_id,
    )

    return (
        cell46_checkpoint_dir
        / f"{safe_id}.json"
    )


# =========================================================
# ATOMIC WRITE
# =========================================================

def cell46_atomic_write_json(
    path,
    payload,
):

    path = Path(path)

    content = (
        json.dumps(
            payload,
            ensure_ascii=False,
            sort_keys=True,
            indent=2,
        )
        + "\n"
    )

    fd, temp_name = tempfile.mkstemp(
        prefix=path.name + ".",
        suffix=".tmp",
        dir=str(path.parent),
    )

    try:

        with os.fdopen(
            fd,
            "w",
            encoding="utf-8",
        ) as handle:

            handle.write(content)
            handle.flush()
            os.fsync(
                handle.fileno()
            )

        os.replace(
            temp_name,
            path,
        )

    except Exception:

        try:
            os.unlink(
                temp_name
            )
        except FileNotFoundError:
            pass

        raise


# =========================================================
# CHECKPOINT CONTRACT
# =========================================================

def cell46_build_checkpoint(
    request,
    result,
):

    checkpoint = {
        "checkpoint_version": (
            CELL46_RUNNER_VERSION
        ),

        "candidate_id": (
            request["candidate_id"]
        ),

        "request_fingerprint": (
            request[
                "request_fingerprint"
            ]
        ),

        "recovery_reason": (
            request[
                "recovery_reason"
            ]
        ),

        "original_request_fingerprint": (
            request[
                "original_request_fingerprint"
            ]
        ),

        "protocol_version": (
            GENERATION_V4_VERSION
        ),

        "protocol_sha256": (
            GENERATION_V4_PROTOCOL_SHA256
        ),

        "prompt_version": (
            GENERATION_V4_PROMPT_VERSION
        ),

        "output_schema_version": (
            GENERATION_V4_OUTPUT_SCHEMA_VERSION
        ),

        "model": (
            semantic_review_model
        ),

        "temperature": (
            CELL46_TEMPERATURE
        ),

        "top_p": (
            CELL46_TOP_P
        ),

        "max_tokens": (
            CELL46_MAX_TOKENS
        ),

        "result": result,
    }

    checkpoint[
        "checkpoint_payload_sha256"
    ] = cell46_sha256_text(
        cell46_canonical_json(
            checkpoint
        )
    )

    return checkpoint


def cell46_validate_checkpoint(
    checkpoint,
    request,
):

    errors = []

    if not isinstance(
        checkpoint,
        dict,
    ):
        return [
            "CHECKPOINT_NOT_OBJECT"
        ]

    stored_sha = checkpoint.get(
        "checkpoint_payload_sha256"
    )

    payload_without_sha = {
        key: value
        for key, value
        in checkpoint.items()
        if key
        != "checkpoint_payload_sha256"
    }

    recalculated_sha = (
        cell46_sha256_text(
            cell46_canonical_json(
                payload_without_sha
            )
        )
    )

    if stored_sha != recalculated_sha:

        errors.append(
            "CHECKPOINT_HASH_MISMATCH"
        )

    exact_expectations = {
        "checkpoint_version": (
            CELL46_RUNNER_VERSION
        ),

        "candidate_id": (
            request["candidate_id"]
        ),

        "request_fingerprint": (
            request[
                "request_fingerprint"
            ]
        ),

        "recovery_reason": (
            request[
                "recovery_reason"
            ]
        ),

        "original_request_fingerprint": (
            request[
                "original_request_fingerprint"
            ]
        ),

        "protocol_version": (
            GENERATION_V4_VERSION
        ),

        "protocol_sha256": (
            GENERATION_V4_PROTOCOL_SHA256
        ),

        "prompt_version": (
            GENERATION_V4_PROMPT_VERSION
        ),

        "output_schema_version": (
            GENERATION_V4_OUTPUT_SCHEMA_VERSION
        ),

        "model": (
            semantic_review_model
        ),

        "temperature": (
            CELL46_TEMPERATURE
        ),

        "top_p": (
            CELL46_TOP_P
        ),

        "max_tokens": (
            CELL46_MAX_TOKENS
        ),
    }

    for (
        field_name,
        expected_value,
    ) in exact_expectations.items():

        if (
            checkpoint.get(
                field_name
            )
            != expected_value
        ):

            errors.append(
                f"{field_name.upper()}_MISMATCH"
            )

    result = checkpoint.get(
        "result"
    )

    if not isinstance(
        result,
        dict,
    ):

        errors.append(
            "CHECKPOINT_RESULT_INVALID"
        )

    elif (
        result.get(
            "candidate_id"
        )
        != request[
            "candidate_id"
        ]
    ):

        errors.append(
            "CHECKPOINT_RESULT_CANDIDATE_MISMATCH"
        )

    return sorted(
        set(errors)
    )


def cell46_load_checkpoint(request):

    path = cell46_checkpoint_path(
        request["candidate_id"]
    )

    if not path.exists():

        return (
            None,
            "MISSING",
            [],
        )

    try:

        with path.open(
            "r",
            encoding="utf-8",
        ) as handle:

            checkpoint = json.load(
                handle
            )

    except Exception as exc:

        return (
            None,
            "UNREADABLE",
            [
                f"{type(exc).__name__}: {exc}"
            ],
        )

    errors = (
        cell46_validate_checkpoint(
            checkpoint,
            request,
        )
    )

    if errors:

        return (
            checkpoint,
            "INVALID",
            errors,
        )

    return (
        checkpoint,
        "VALID",
        [],
    )


# =========================================================
# COST
# =========================================================

cell46_has_cost_rates = (
    "INPUT_TOMAN_PER_TOKEN" in globals()
    and
    "OUTPUT_TOMAN_PER_TOKEN" in globals()
)


def cell46_estimated_cost_toman(
    input_tokens,
    output_tokens,
):

    if not cell46_has_cost_rates:
        return None

    return (
        input_tokens
        * INPUT_TOMAN_PER_TOKEN
        +
        output_tokens
        * OUTPUT_TOMAN_PER_TOKEN
    )


# =========================================================
# API EXECUTION
# =========================================================

def cell46_execute_request(request):

    attempt_history = []
    api_attempts = 0

    for attempt_number in range(
        1,
        CELL46_MAX_ATTEMPTS + 1,
    ):

        started_at = time.time()
        api_attempts += 1

        try:

            response = (
                semantic_review_client
                .chat
                .completions
                .create(
                    model=(
                        semantic_review_model
                    ),

                    messages=[
                        {
                            "role": "system",
                            "content": (
                                request[
                                    "system_prompt"
                                ]
                            ),
                        },
                        {
                            "role": "user",
                            "content": (
                                request[
                                    "user_prompt"
                                ]
                            ),
                        },
                    ],

                    temperature=(
                        CELL46_TEMPERATURE
                    ),

                    top_p=(
                        CELL46_TOP_P
                    ),

                    max_tokens=(
                        CELL46_MAX_TOKENS
                    ),

                    extra_body={
                        "enable_thinking": False,
                    },
                )
            )

        except Exception as exc:

            attempt_history.append({
                "attempt": (
                    attempt_number
                ),

                "status": (
                    "API_ERROR"
                ),

                "error": (
                    f"{type(exc).__name__}: {exc}"
                ),
            })

            if (
                attempt_number
                < CELL46_MAX_ATTEMPTS
            ):

                wait_seconds = (
                    70.0
                    if cell46_is_rate_limit_exception(
                        exc
                    )
                    else (
                        CELL46_RETRY_SECONDS
                        * attempt_number
                    )
                )

                time.sleep(
                    wait_seconds
                )

                continue

            return {
                "candidate_id": (
                    request["candidate_id"]
                ),

                "request_fingerprint": (
                    request[
                        "request_fingerprint"
                    ]
                ),

                "protocol_version": (
                    GENERATION_V4_VERSION
                ),

                "protocol_sha256": (
                    GENERATION_V4_PROTOCOL_SHA256
                ),

                "recovery_reason": (
                    request[
                        "recovery_reason"
                    ]
                ),

                "execution_status": (
                    "API_FAILED"
                ),

                "api_attempts": (
                    api_attempts
                ),

                "attempt_history": (
                    attempt_history
                ),

                "structural_status": (
                    "FAIL"
                ),

                "structural_errors": [
                    "API_FAILED"
                ],
            }

        elapsed_seconds = (
            time.time()
            - started_at
        )

        choice = response.choices[0]

        finish_reason = getattr(
            choice,
            "finish_reason",
            None,
        )

        raw_content = getattr(
            choice.message,
            "content",
            None,
        )

        usage = getattr(
            response,
            "usage",
            None,
        )

        input_tokens = int(
            getattr(
                usage,
                "prompt_tokens",
                0,
            )
            or 0
        )

        output_tokens = int(
            getattr(
                usage,
                "completion_tokens",
                0,
            )
            or 0
        )

        total_tokens = int(
            getattr(
                usage,
                "total_tokens",
                0,
            )
            or (
                input_tokens
                + output_tokens
            )
        )

        reasoning_tokens = 0

        completion_details = getattr(
            usage,
            "completion_tokens_details",
            None,
        )

        if completion_details is not None:

            reasoning_tokens = int(
                getattr(
                    completion_details,
                    "reasoning_tokens",
                    0,
                )
                or 0
            )

        clean_content = (
            cell46_clean_model_json(
                raw_content
            )
        )

        parsed_json = None
        parse_error = None

        try:

            parsed_json = json.loads(
                clean_content
            )

        except Exception as exc:

            parse_error = (
                f"{type(exc).__name__}: {exc}"
            )

        if parsed_json is None:

            structural_errors = [
                "JSON_PARSE_FAILED"
            ]

        else:

            structural_errors = (
                cell41_validate_generation_output_v4(
                    parsed_json
                )
            )

        structural_status = (
            "PASS"
            if not structural_errors
            else "FAIL"
        )

        return {
            "candidate_id": (
                request["candidate_id"]
            ),

            "request_fingerprint": (
                request[
                    "request_fingerprint"
                ]
            ),

            "protocol_version": (
                GENERATION_V4_VERSION
            ),

            "protocol_sha256": (
                GENERATION_V4_PROTOCOL_SHA256
            ),

            "recovery_reason": (
                request[
                    "recovery_reason"
                ]
            ),

            "execution_status": (
                "COMPLETED"
            ),

            "api_attempts": (
                api_attempts
            ),

            "finish_reason": (
                finish_reason
            ),

            "elapsed_seconds": round(
                elapsed_seconds,
                4,
            ),

            "usage": {
                "input_tokens": (
                    input_tokens
                ),

                "output_tokens": (
                    output_tokens
                ),

                "reasoning_tokens": (
                    reasoning_tokens
                ),

                "total_tokens": (
                    total_tokens
                ),
            },

            "estimated_cost_toman": (
                cell46_estimated_cost_toman(
                    input_tokens,
                    output_tokens,
                )
            ),

            "raw_content": (
                raw_content
            ),

            "parsed_json": (
                parsed_json
            ),

            "parse_error": (
                parse_error
            ),

            "structural_errors": (
                structural_errors
            ),

            "structural_status": (
                structural_status
            ),

            "attempt_history": (
                attempt_history
            ),
        }

    raise RuntimeError(
        "Unexpected Cell 46 execution path."
    )


# =========================================================
# PREFLIGHT SCAN
# =========================================================

cell46_valid_before_api = {}
cell46_missing_requests = []
cell46_invalid_before_api = []


for request in cell46_requests:

    (
        checkpoint,
        status,
        errors,
    ) = cell46_load_checkpoint(
        request
    )

    candidate_id = request[
        "candidate_id"
    ]

    if status == "VALID":

        cell46_valid_before_api[
            candidate_id
        ] = checkpoint

    elif status == "MISSING":

        cell46_missing_requests.append(
            request
        )

    else:

        cell46_invalid_before_api.append({
            "candidate_id": (
                candidate_id
            ),

            "status": (
                status
            ),

            "errors": (
                errors
            ),
        })


assert not cell46_invalid_before_api, (
    "Invalid/unreadable Cell 46 checkpoint detected. "
    "Refusing automatic paid replacement. "
    f"Examples: {cell46_invalid_before_api[:5]}"
)


# =========================================================
# PREFLIGHT OUTPUT
# =========================================================

print(
    "=== GOLDEN V2 TARGETED FULL GENERATION "
    "RECOVERY PREFLIGHT ==="
)

print()

print(
    "Recovery population:",
    len(cell46_requests),
)

print(
    "Length failures:",
    len(
        cell45_length_recovery_candidate_ids
    ),
)

print(
    "API failures:",
    len(
        cell45_api_recovery_candidate_ids
    ),
)

print(
    "Exact valid recovery checkpoints:",
    len(cell46_valid_before_api),
)

print(
    "Missing recovery checkpoints:",
    len(cell46_missing_requests),
)

print(
    "Successful Cell 44 outputs preserved:",
    len(cell45_pass_results),
)

print(
    "Model:",
    semantic_review_model,
)

print(
    "Protocol:",
    GENERATION_V4_VERSION,
)

print(
    "Recovery max tokens:",
    CELL46_MAX_TOKENS,
)

print(
    "Parallel workers:",
    CELL46_MAX_WORKERS,
)

print(
    "API allowed for missing exact checkpoints:",
    CELL46_ALLOW_API_FOR_MISSING,
)

print()


if (
    cell46_missing_requests
    and not CELL46_ALLOW_API_FOR_MISSING
):

    raise AssertionError(
        "Cell 46 has missing recovery checkpoints "
        "but API execution is disabled."
    )


# =========================================================
# TARGETED RECOVERY EXECUTION
# =========================================================

cell46_api_candidates_this_execution = 0
cell46_api_attempts_this_execution = 0
cell46_checkpoints_created_this_execution = 0


def cell46_execute_and_checkpoint(
    request,
):

    candidate_id = request[
        "candidate_id"
    ]

    result = (
        cell46_execute_request(
            request
        )
    )

    checkpoint = (
        cell46_build_checkpoint(
            request,
            result,
        )
    )

    cell46_atomic_write_json(
        cell46_checkpoint_path(
            candidate_id
        ),
        checkpoint,
    )

    (
        verified_checkpoint,
        verified_status,
        verified_errors,
    ) = cell46_load_checkpoint(
        request
    )

    if verified_status != "VALID":

        raise AssertionError(
            "Cell 46 checkpoint verification failed "
            f"for {candidate_id}: "
            f"{verified_errors}"
        )

    return (
        candidate_id,
        verified_checkpoint[
            "result"
        ],
    )


if cell46_missing_requests:

    completed_count = 0

    with ThreadPoolExecutor(
        max_workers=CELL46_MAX_WORKERS
    ) as executor:

        future_to_request = {
            executor.submit(
                cell46_execute_and_checkpoint,
                request,
            ): request
            for request
            in cell46_missing_requests
        }

        for future in as_completed(
            future_to_request
        ):

            request = (
                future_to_request[
                    future
                ]
            )

            candidate_id = request[
                "candidate_id"
            ]

            try:

                (
                    resolved_candidate_id,
                    result,
                ) = future.result()

            except Exception as exc:

                raise RuntimeError(
                    "Cell 46 worker/checkpoint failure for "
                    f"{candidate_id}: "
                    f"{type(exc).__name__}: {exc}"
                ) from exc

            assert (
                resolved_candidate_id
                == candidate_id
            )

            completed_count += 1

            cell46_api_candidates_this_execution += 1

            cell46_api_attempts_this_execution += int(
                result.get(
                    "api_attempts",
                    0,
                )
                or 0
            )

            cell46_checkpoints_created_this_execution += 1

            print(
                f"[{completed_count:03d}/"
                f"{len(cell46_missing_requests):03d}]",
                candidate_id,
                "| reason=",
                request[
                    "recovery_reason"
                ],
                "| structural=",
                result.get(
                    "structural_status"
                ),
                "| finish=",
                result.get(
                    "finish_reason"
                ),
                flush=True,
            )


# =========================================================
# FINAL RECOVERY DISK RESCAN
# =========================================================

cell46_recovery_results = []

cell46_valid_checkpoint_count = 0
cell46_missing_checkpoint_count = 0
cell46_invalid_checkpoint_count = 0


for request in cell46_requests:

    (
        checkpoint,
        status,
        errors,
    ) = cell46_load_checkpoint(
        request
    )

    if status == "VALID":

        cell46_valid_checkpoint_count += 1

        cell46_recovery_results.append(
            checkpoint[
                "result"
            ]
        )

    elif status == "MISSING":

        cell46_missing_checkpoint_count += 1

    else:

        cell46_invalid_checkpoint_count += 1


# =========================================================
# RECOVERY RESULT COUNTS
# =========================================================

cell46_structural_distribution = Counter(
    result.get(
        "structural_status"
    )
    for result
    in cell46_recovery_results
)

cell46_finish_reason_distribution = Counter(
    result.get(
        "finish_reason"
    )
    for result
    in cell46_recovery_results
)

cell46_execution_status_distribution = Counter(
    result.get(
        "execution_status"
    )
    for result
    in cell46_recovery_results
)


cell46_structural_pass_count = (
    cell46_structural_distribution.get(
        "PASS",
        0,
    )
)

cell46_structural_fail_count = (
    cell46_structural_distribution.get(
        "FAIL",
        0,
    )
)

cell46_api_failure_count = (
    cell46_execution_status_distribution.get(
        "API_FAILED",
        0,
    )
)


# =========================================================
# COST
# =========================================================

cell46_known_costs = [
    result.get(
        "estimated_cost_toman"
    )
    for result
    in cell46_recovery_results
    if isinstance(
        result.get(
            "estimated_cost_toman"
        ),
        (int, float),
    )
]

cell46_total_estimated_cost_toman = (
    sum(cell46_known_costs)
    if cell46_known_costs
    else None
)


# =========================================================
# RECONSTRUCT FINAL 3813 GENERATION RESULTS
# =========================================================

cell46_recovery_result_by_candidate_id = {
    result["candidate_id"]: result
    for result
    in cell46_recovery_results
}

assert len(
    cell46_recovery_result_by_candidate_id
) == len(
    cell46_recovery_results
)


cell46_final_generation_results = []


for original_result in cell44_full_generation_results:

    candidate_id = (
        original_result[
            "candidate_id"
        ]
    )

    if (
        candidate_id
        in cell46_recovery_result_by_candidate_id
    ):

        cell46_final_generation_results.append(
            cell46_recovery_result_by_candidate_id[
                candidate_id
            ]
        )

    else:

        cell46_final_generation_results.append(
            original_result
        )


assert len(
    cell46_final_generation_results
) == 3813

assert len({
    result["candidate_id"]
    for result
    in cell46_final_generation_results
}) == 3813


# =========================================================
# FINAL 3813 STATE
# =========================================================

cell46_final_structural_distribution = Counter(
    result.get(
        "structural_status"
    )
    for result
    in cell46_final_generation_results
)

cell46_final_execution_status_distribution = Counter(
    result.get(
        "execution_status"
    )
    for result
    in cell46_final_generation_results
)

cell46_final_finish_reason_distribution = Counter(
    result.get(
        "finish_reason"
    )
    for result
    in cell46_final_generation_results
)


cell46_final_structural_pass_count = (
    cell46_final_structural_distribution.get(
        "PASS",
        0,
    )
)

cell46_final_structural_fail_count = (
    cell46_final_structural_distribution.get(
        "FAIL",
        0,
    )
)

cell46_final_api_failure_count = (
    cell46_final_execution_status_distribution.get(
        "API_FAILED",
        0,
    )
)


# =========================================================
# RUNNER HEALTH
# =========================================================

cell46_runner_checks = {
    "182_recovery_requests": (
        len(cell46_requests)
        == 182
    ),

    "182_recovery_results_reconstructed": (
        len(cell46_recovery_results)
        == 182
    ),

    "182_valid_recovery_checkpoints": (
        cell46_valid_checkpoint_count
        == 182
    ),

    "zero_missing_recovery_checkpoints": (
        cell46_missing_checkpoint_count
        == 0
    ),

    "zero_invalid_recovery_checkpoints": (
        cell46_invalid_checkpoint_count
        == 0
    ),

    "final_population_3813": (
        len(
            cell46_final_generation_results
        )
        == 3813
    ),

    "final_unique_candidates_3813": (
        len({
            result["candidate_id"]
            for result
            in cell46_final_generation_results
        })
        == 3813
    ),

    "3631_original_successes_preserved": (
        all(
            candidate_id
            not in set(
                cell45_all_recovery_candidate_ids
            )
            for candidate_id in [
                row["candidate_id"]
                for row
                in cell45_pass_results
            ]
        )
    ),
}


cell46_runner_status = (
    "PASS"
    if all(
        cell46_runner_checks.values()
    )
    else "FAIL"
)


# Experimental outcome kept separate from runner health.
cell46_recovery_structural_status = (
    "PASS"
    if (
        cell46_structural_pass_count
        == 182
        and
        cell46_api_failure_count
        == 0
    )
    else "FAIL"
)


cell46_final_generation_status = (
    "PASS"
    if (
        cell46_final_structural_pass_count
        == 3813
        and
        cell46_final_structural_fail_count
        == 0
        and
        cell46_final_api_failure_count
        == 0
    )
    else "FAIL"
)


if cell46_runner_status != "PASS":

    cell46_next_readiness = (
        "BLOCKED_CELL46_RUNNER_REPAIR_REQUIRED"
    )

elif cell46_api_failure_count > 0:

    cell46_next_readiness = (
        "READY_FOR_CELL46_API_FAILURE_RECOVERY"
    )

elif cell46_structural_fail_count > 0:

    cell46_next_readiness = (
        "READY_FOR_CELL46_STRUCTURAL_FAILURE_ANALYSIS"
    )

elif cell46_final_generation_status == "PASS":

    cell46_next_readiness = (
        "READY_FOR_FULL_QA_SEMANTIC_VALIDATION_V4"
    )

else:

    cell46_next_readiness = (
        "READY_FOR_FINAL_GENERATION_STATE_ANALYSIS"
    )


# =========================================================
# OUTPUT
# =========================================================

print()

print(
    "=== GOLDEN V2 TARGETED FULL GENERATION "
    "RECOVERY RESULT ==="
)

print()

print(
    "Recovery population:",
    len(cell46_requests),
)

print(
    "Recovery results reconstructed:",
    len(cell46_recovery_results),
)

print(
    "Valid recovery checkpoints:",
    cell46_valid_checkpoint_count,
)

print(
    "Missing recovery checkpoints:",
    cell46_missing_checkpoint_count,
)

print(
    "Invalid recovery checkpoints:",
    cell46_invalid_checkpoint_count,
)

print()

print(
    "API candidates this execution:",
    cell46_api_candidates_this_execution,
)

print(
    "API attempts this execution:",
    cell46_api_attempts_this_execution,
)

print(
    "Recovery checkpoints created this execution:",
    cell46_checkpoints_created_this_execution,
)

print()

print(
    "Recovery structural distribution:",
    dict(
        cell46_structural_distribution
    ),
)

print(
    "Recovery finish reasons:",
    dict(
        cell46_finish_reason_distribution
    ),
)

print(
    "Recovery execution statuses:",
    dict(
        cell46_execution_status_distribution
    ),
)

print(
    "Recovery API failures:",
    cell46_api_failure_count,
)

print(
    "Estimated recovery cost represented by checkpoints (toman):",
    cell46_total_estimated_cost_toman,
)

print()

print(
    "=== FINAL 3813 GENERATION STATE ==="
)

print(
    "Population:",
    len(
        cell46_final_generation_results
    ),
)

print(
    "Structural distribution:",
    dict(
        cell46_final_structural_distribution
    ),
)

print(
    "Structural PASS:",
    cell46_final_structural_pass_count,
    "/ 3813",
)

print(
    "Structural FAIL:",
    cell46_final_structural_fail_count,
    "/ 3813",
)

print(
    "API failures:",
    cell46_final_api_failure_count,
)

print(
    "Finish reason distribution:",
    dict(
        cell46_final_finish_reason_distribution
    ),
)

print()

print(
    "Runner checks:"
)

for (
    check_name,
    check_value,
) in cell46_runner_checks.items():

    print(
        " -",
        check_name,
        ":",
        (
            "PASS"
            if check_value
            else "FAIL"
        ),
    )

print()

print(
    "Runner health:",
    cell46_runner_status,
)

print(
    "Recovery structural result:",
    cell46_recovery_structural_status,
)

print(
    "Final Full Generation result:",
    cell46_final_generation_status,
)

print(
    "Next readiness:",
    cell46_next_readiness,
)

print(
    "CELL 46 CHECKPOINTED RUNNER:",
    cell46_runner_status,
)


assert cell46_runner_status == "PASS", (
    "Cell 46 checkpointed runner health failed."
)
````

پس از Recovery مرحله قبل، ۱۸۰ مورد از ۱۸۲ شکست اصلاح شدند و تنها دو خروجی به علت رسیدن دوباره به محدودیت طول ۳۰۷۲ توکن نامعتبر باقی ماندند.
در این مرحله فقط همین دو Candidate با Fingerprint و Checkpoint مستقل و سقف خروجی ۶۱۴۴ توکن دوباره اجرا می‌شوند.
تمام ۳۸۱۱ خروجی ساختاری سالم قبلی بدون تغییر حفظ می‌شوند.
پس از بازیابی، جمعیت کامل ۳۸۱۳ خروجی بازسازی می‌شود و تنها در صورت Structural PASS کامل، مسیر برای Semantic Validation سراسری باز خواهد شد.

````python
# Code Cell 47 — Final Two-Case Full Generation Recovery

from collections import Counter
from concurrent.futures import ThreadPoolExecutor, as_completed
from pathlib import Path
import hashlib
import json
import os
import re
import tempfile
import time


# =========================================================
# CONFIG
# =========================================================

CELL47_RUNNER_VERSION = (
    "golden_v2_full_generation_v4_final_length_recovery_v1"
)

CELL47_MAX_TOKENS = 6144
CELL47_TEMPERATURE = 0.7
CELL47_TOP_P = 0.8

CELL47_MAX_ATTEMPTS = 2
CELL47_RETRY_SECONDS = 2.0
CELL47_MAX_WORKERS = 2

# Intentional paid execution for ONLY the remaining 2 cases.
# MUST become False before final Restart -> Run All.
CELL47_ALLOW_API_FOR_MISSING = False


# =========================================================
# REQUIRED STATE
# =========================================================

required_cell47_names = [
    "project_root",
    "semantic_review_client",
    "semantic_review_model",
    "cell46_final_generation_results",
    "cell46_recovery_results",
    "cell44_request_by_candidate_id",
    "cell41_validate_generation_output_v4",
    "GOLDEN_V2_GENERATION_SYSTEM_PROMPT_V4",
    "GENERATION_V4_VERSION",
    "GENERATION_V4_PROMPT_VERSION",
    "GENERATION_V4_OUTPUT_SCHEMA_VERSION",
    "GENERATION_V4_PROTOCOL_SHA256",
]

missing_cell47_names = [
    name
    for name in required_cell47_names
    if name not in globals()
]

assert not missing_cell47_names, (
    "Missing required Cell 47 state: "
    + ", ".join(missing_cell47_names)
)

assert len(cell46_final_generation_results) == 3813


# =========================================================
# EXACT TWO REMAINING FAILURES
# =========================================================

cell47_failed_results = [
    result
    for result in cell46_final_generation_results
    if result.get("structural_status") == "FAIL"
]

assert len(cell47_failed_results) == 2, (
    "Cell 47 expected exactly 2 remaining structural failures, "
    f"found {len(cell47_failed_results)}."
)

assert all(
    result.get("finish_reason") == "length"
    for result in cell47_failed_results
), (
    "Cell 47 expected both remaining failures "
    "to be finish_reason=length."
)

cell47_candidate_ids = [
    result["candidate_id"]
    for result in cell47_failed_results
]

assert len(set(cell47_candidate_ids)) == 2


# =========================================================
# HASH UTILITIES
# =========================================================

def cell47_canonical_json(value):

    return json.dumps(
        value,
        ensure_ascii=False,
        sort_keys=True,
        separators=(",", ":"),
    )


def cell47_sha256_text(text):

    return hashlib.sha256(
        text.encode("utf-8")
    ).hexdigest()


# =========================================================
# OUTPUT CLEANING
# =========================================================

def cell47_clean_model_json(content):

    if not isinstance(content, str):
        return ""

    clean = content.strip()

    if clean.startswith("```"):

        clean = re.sub(
            r"^```(?:json)?\s*",
            "",
            clean,
            flags=re.IGNORECASE,
        )

        clean = re.sub(
            r"\s*```$",
            "",
            clean,
        )

    return clean.strip()


def cell47_is_rate_limit_exception(exc):

    error_text = str(exc).lower()
    error_type = type(exc).__name__.lower()

    return (
        "ratelimit" in error_type
        or "rate_limit" in error_type
        or "429" in error_text
        or "too many requests" in error_text
    )


# =========================================================
# REQUEST CONTRACT
# =========================================================

cell47_previous_failure_by_candidate_id = {
    result["candidate_id"]: result
    for result in cell47_failed_results
}


def cell47_build_request(candidate_id):

    original_request = (
        cell44_request_by_candidate_id[
            candidate_id
        ]
    )

    previous_failure = (
        cell47_previous_failure_by_candidate_id[
            candidate_id
        ]
    )

    previous_result_hash = (
        cell47_sha256_text(
            cell47_canonical_json(
                previous_failure
            )
        )
    )

    contract = {
        "runner_version": (
            CELL47_RUNNER_VERSION
        ),

        "candidate_id": candidate_id,

        "recovery_reason": (
            "SECOND_LENGTH_RECOVERY"
        ),

        "previous_result_sha256": (
            previous_result_hash
        ),

        "original_request_fingerprint": (
            original_request[
                "request_fingerprint"
            ]
        ),

        "protocol_version": (
            GENERATION_V4_VERSION
        ),

        "protocol_sha256": (
            GENERATION_V4_PROTOCOL_SHA256
        ),

        "prompt_version": (
            GENERATION_V4_PROMPT_VERSION
        ),

        "output_schema_version": (
            GENERATION_V4_OUTPUT_SCHEMA_VERSION
        ),

        "model": (
            semantic_review_model
        ),

        "system_prompt": (
            GOLDEN_V2_GENERATION_SYSTEM_PROMPT_V4
        ),

        "user_prompt": (
            original_request[
                "user_prompt"
            ]
        ),

        "temperature": (
            CELL47_TEMPERATURE
        ),

        "top_p": (
            CELL47_TOP_P
        ),

        "max_tokens": (
            CELL47_MAX_TOKENS
        ),

        "enable_thinking": False,
    }

    request_fingerprint = (
        cell47_sha256_text(
            cell47_canonical_json(
                contract
            )
        )
    )

    return {
        **contract,
        "request_fingerprint": (
            request_fingerprint
        ),
    }


cell47_requests = [
    cell47_build_request(
        candidate_id
    )
    for candidate_id in cell47_candidate_ids
]

assert len(cell47_requests) == 2

assert len({
    request["request_fingerprint"]
    for request in cell47_requests
}) == 2


# =========================================================
# CHECKPOINT DIRECTORY
# =========================================================

cell47_checkpoint_dir = (
    Path(project_root)
    / "artifacts"
    / "phase2"
    / "golden_dataset_v2"
    / "work"
    / "full_qa_generation_v4_final_length_recovery_6144"
)

cell47_checkpoint_dir.mkdir(
    parents=True,
    exist_ok=True,
)


def cell47_checkpoint_path(candidate_id):

    safe_id = re.sub(
        r"[^A-Za-z0-9_.-]+",
        "_",
        candidate_id,
    )

    return (
        cell47_checkpoint_dir
        / f"{safe_id}.json"
    )


# =========================================================
# ATOMIC WRITE
# =========================================================

def cell47_atomic_write_json(path, payload):

    path = Path(path)

    content = (
        json.dumps(
            payload,
            ensure_ascii=False,
            sort_keys=True,
            indent=2,
        )
        + "\n"
    )

    fd, temp_name = tempfile.mkstemp(
        prefix=path.name + ".",
        suffix=".tmp",
        dir=str(path.parent),
    )

    try:

        with os.fdopen(
            fd,
            "w",
            encoding="utf-8",
        ) as handle:

            handle.write(content)
            handle.flush()
            os.fsync(handle.fileno())

        os.replace(
            temp_name,
            path,
        )

    except Exception:

        try:
            os.unlink(temp_name)
        except FileNotFoundError:
            pass

        raise


# =========================================================
# CHECKPOINT CONTRACT
# =========================================================

def cell47_build_checkpoint(request, result):

    checkpoint = {
        "checkpoint_version": (
            CELL47_RUNNER_VERSION
        ),

        "candidate_id": (
            request["candidate_id"]
        ),

        "request_fingerprint": (
            request["request_fingerprint"]
        ),

        "previous_result_sha256": (
            request["previous_result_sha256"]
        ),

        "protocol_version": (
            GENERATION_V4_VERSION
        ),

        "protocol_sha256": (
            GENERATION_V4_PROTOCOL_SHA256
        ),

        "model": (
            semantic_review_model
        ),

        "temperature": (
            CELL47_TEMPERATURE
        ),

        "top_p": (
            CELL47_TOP_P
        ),

        "max_tokens": (
            CELL47_MAX_TOKENS
        ),

        "result": result,
    }

    checkpoint[
        "checkpoint_payload_sha256"
    ] = cell47_sha256_text(
        cell47_canonical_json(
            checkpoint
        )
    )

    return checkpoint


def cell47_validate_checkpoint(
    checkpoint,
    request,
):

    errors = []

    if not isinstance(checkpoint, dict):
        return ["CHECKPOINT_NOT_OBJECT"]

    stored_sha = checkpoint.get(
        "checkpoint_payload_sha256"
    )

    payload_without_sha = {
        key: value
        for key, value in checkpoint.items()
        if key != "checkpoint_payload_sha256"
    }

    recalculated_sha = (
        cell47_sha256_text(
            cell47_canonical_json(
                payload_without_sha
            )
        )
    )

    if stored_sha != recalculated_sha:
        errors.append(
            "CHECKPOINT_HASH_MISMATCH"
        )

    exact_expectations = {
        "checkpoint_version": (
            CELL47_RUNNER_VERSION
        ),

        "candidate_id": (
            request["candidate_id"]
        ),

        "request_fingerprint": (
            request["request_fingerprint"]
        ),

        "previous_result_sha256": (
            request["previous_result_sha256"]
        ),

        "protocol_version": (
            GENERATION_V4_VERSION
        ),

        "protocol_sha256": (
            GENERATION_V4_PROTOCOL_SHA256
        ),

        "model": (
            semantic_review_model
        ),

        "temperature": (
            CELL47_TEMPERATURE
        ),

        "top_p": (
            CELL47_TOP_P
        ),

        "max_tokens": (
            CELL47_MAX_TOKENS
        ),
    }

    for field_name, expected_value in (
        exact_expectations.items()
    ):

        if (
            checkpoint.get(field_name)
            != expected_value
        ):

            errors.append(
                f"{field_name.upper()}_MISMATCH"
            )

    result = checkpoint.get("result")

    if not isinstance(result, dict):

        errors.append(
            "CHECKPOINT_RESULT_INVALID"
        )

    elif (
        result.get("candidate_id")
        != request["candidate_id"]
    ):

        errors.append(
            "CHECKPOINT_RESULT_CANDIDATE_MISMATCH"
        )

    return sorted(set(errors))


def cell47_load_checkpoint(request):

    path = cell47_checkpoint_path(
        request["candidate_id"]
    )

    if not path.exists():

        return (
            None,
            "MISSING",
            [],
        )

    try:

        with path.open(
            "r",
            encoding="utf-8",
        ) as handle:

            checkpoint = json.load(handle)

    except Exception as exc:

        return (
            None,
            "UNREADABLE",
            [
                f"{type(exc).__name__}: {exc}"
            ],
        )

    errors = cell47_validate_checkpoint(
        checkpoint,
        request,
    )

    if errors:

        return (
            checkpoint,
            "INVALID",
            errors,
        )

    return (
        checkpoint,
        "VALID",
        [],
    )


# =========================================================
# COST
# =========================================================

cell47_has_cost_rates = (
    "INPUT_TOMAN_PER_TOKEN" in globals()
    and
    "OUTPUT_TOMAN_PER_TOKEN" in globals()
)


def cell47_estimated_cost_toman(
    input_tokens,
    output_tokens,
):

    if not cell47_has_cost_rates:
        return None

    return (
        input_tokens
        * INPUT_TOMAN_PER_TOKEN
        +
        output_tokens
        * OUTPUT_TOMAN_PER_TOKEN
    )


# =========================================================
# API EXECUTION
# =========================================================

def cell47_execute_request(request):

    attempt_history = []
    api_attempts = 0

    for attempt_number in range(
        1,
        CELL47_MAX_ATTEMPTS + 1,
    ):

        started_at = time.time()
        api_attempts += 1

        try:

            response = (
                semantic_review_client
                .chat
                .completions
                .create(
                    model=semantic_review_model,

                    messages=[
                        {
                            "role": "system",
                            "content": (
                                request[
                                    "system_prompt"
                                ]
                            ),
                        },
                        {
                            "role": "user",
                            "content": (
                                request[
                                    "user_prompt"
                                ]
                            ),
                        },
                    ],

                    temperature=(
                        CELL47_TEMPERATURE
                    ),

                    top_p=(
                        CELL47_TOP_P
                    ),

                    max_tokens=(
                        CELL47_MAX_TOKENS
                    ),

                    extra_body={
                        "enable_thinking": False,
                    },
                )
            )

        except Exception as exc:

            attempt_history.append({
                "attempt": attempt_number,
                "status": "API_ERROR",
                "error": (
                    f"{type(exc).__name__}: {exc}"
                ),
            })

            if (
                attempt_number
                < CELL47_MAX_ATTEMPTS
            ):

                wait_seconds = (
                    70.0
                    if cell47_is_rate_limit_exception(
                        exc
                    )
                    else (
                        CELL47_RETRY_SECONDS
                        * attempt_number
                    )
                )

                time.sleep(wait_seconds)
                continue

            return {
                "candidate_id": (
                    request["candidate_id"]
                ),

                "request_fingerprint": (
                    request[
                        "request_fingerprint"
                    ]
                ),

                "protocol_version": (
                    GENERATION_V4_VERSION
                ),

                "protocol_sha256": (
                    GENERATION_V4_PROTOCOL_SHA256
                ),

                "execution_status": (
                    "API_FAILED"
                ),

                "api_attempts": (
                    api_attempts
                ),

                "attempt_history": (
                    attempt_history
                ),

                "structural_status": (
                    "FAIL"
                ),

                "structural_errors": [
                    "API_FAILED"
                ],
            }

        elapsed_seconds = (
            time.time()
            - started_at
        )

        choice = response.choices[0]

        finish_reason = getattr(
            choice,
            "finish_reason",
            None,
        )

        raw_content = getattr(
            choice.message,
            "content",
            None,
        )

        usage = getattr(
            response,
            "usage",
            None,
        )

        input_tokens = int(
            getattr(
                usage,
                "prompt_tokens",
                0,
            )
            or 0
        )

        output_tokens = int(
            getattr(
                usage,
                "completion_tokens",
                0,
            )
            or 0
        )

        total_tokens = int(
            getattr(
                usage,
                "total_tokens",
                0,
            )
            or (
                input_tokens
                + output_tokens
            )
        )

        reasoning_tokens = 0

        completion_details = getattr(
            usage,
            "completion_tokens_details",
            None,
        )

        if completion_details is not None:

            reasoning_tokens = int(
                getattr(
                    completion_details,
                    "reasoning_tokens",
                    0,
                )
                or 0
            )

        clean_content = (
            cell47_clean_model_json(
                raw_content
            )
        )

        parsed_json = None
        parse_error = None

        try:

            parsed_json = json.loads(
                clean_content
            )

        except Exception as exc:

            parse_error = (
                f"{type(exc).__name__}: {exc}"
            )

        if parsed_json is None:

            structural_errors = [
                "JSON_PARSE_FAILED"
            ]

        else:

            structural_errors = (
                cell41_validate_generation_output_v4(
                    parsed_json
                )
            )

        structural_status = (
            "PASS"
            if not structural_errors
            else "FAIL"
        )

        return {
            "candidate_id": (
                request["candidate_id"]
            ),

            "request_fingerprint": (
                request[
                    "request_fingerprint"
                ]
            ),

            "protocol_version": (
                GENERATION_V4_VERSION
            ),

            "protocol_sha256": (
                GENERATION_V4_PROTOCOL_SHA256
            ),

            "execution_status": (
                "COMPLETED"
            ),

            "api_attempts": (
                api_attempts
            ),

            "finish_reason": (
                finish_reason
            ),

            "elapsed_seconds": round(
                elapsed_seconds,
                4,
            ),

            "usage": {
                "input_tokens": (
                    input_tokens
                ),

                "output_tokens": (
                    output_tokens
                ),

                "reasoning_tokens": (
                    reasoning_tokens
                ),

                "total_tokens": (
                    total_tokens
                ),
            },

            "estimated_cost_toman": (
                cell47_estimated_cost_toman(
                    input_tokens,
                    output_tokens,
                )
            ),

            "raw_content": (
                raw_content
            ),

            "parsed_json": (
                parsed_json
            ),

            "parse_error": (
                parse_error
            ),

            "structural_errors": (
                structural_errors
            ),

            "structural_status": (
                structural_status
            ),

            "attempt_history": (
                attempt_history
            ),
        }

    raise RuntimeError(
        "Unexpected Cell 47 execution path."
    )


# =========================================================
# PREFLIGHT SCAN
# =========================================================

cell47_valid_before_api = {}
cell47_missing_requests = []
cell47_invalid_before_api = []


for request in cell47_requests:

    checkpoint, status, errors = (
        cell47_load_checkpoint(
            request
        )
    )

    candidate_id = request[
        "candidate_id"
    ]

    if status == "VALID":

        cell47_valid_before_api[
            candidate_id
        ] = checkpoint

    elif status == "MISSING":

        cell47_missing_requests.append(
            request
        )

    else:

        cell47_invalid_before_api.append({
            "candidate_id": candidate_id,
            "status": status,
            "errors": errors,
        })


assert not cell47_invalid_before_api, (
    "Invalid/unreadable Cell 47 checkpoint detected. "
    "Refusing automatic paid replacement. "
    f"{cell47_invalid_before_api}"
)


# =========================================================
# PREFLIGHT OUTPUT
# =========================================================

print(
    "=== GOLDEN V2 FINAL TWO-CASE "
    "GENERATION RECOVERY PREFLIGHT ==="
)

print()

print(
    "Recovery population:",
    len(cell47_requests),
)

print(
    "Candidates:",
    cell47_candidate_ids,
)

print(
    "Exact valid checkpoints:",
    len(cell47_valid_before_api),
)

print(
    "Missing checkpoints:",
    len(cell47_missing_requests),
)

print(
    "Preserved structural PASS population:",
    3811,
)

print(
    "Model:",
    semantic_review_model,
)

print(
    "Protocol:",
    GENERATION_V4_VERSION,
)

print(
    "Recovery max tokens:",
    CELL47_MAX_TOKENS,
)

print(
    "API allowed for missing exact checkpoints:",
    CELL47_ALLOW_API_FOR_MISSING,
)

print()


if (
    cell47_missing_requests
    and not CELL47_ALLOW_API_FOR_MISSING
):

    raise AssertionError(
        "Cell 47 has missing checkpoints "
        "but API execution is disabled."
    )


# =========================================================
# EXECUTE ONLY MISSING TWO-CASE RECOVERY
# =========================================================

cell47_api_candidates_this_execution = 0
cell47_api_attempts_this_execution = 0
cell47_checkpoints_created_this_execution = 0


def cell47_execute_and_checkpoint(
    request,
):

    result = cell47_execute_request(
        request
    )

    checkpoint = (
        cell47_build_checkpoint(
            request,
            result,
        )
    )

    path = cell47_checkpoint_path(
        request["candidate_id"]
    )

    cell47_atomic_write_json(
        path,
        checkpoint,
    )

    (
        verified_checkpoint,
        verified_status,
        verified_errors,
    ) = cell47_load_checkpoint(
        request
    )

    if verified_status != "VALID":

        raise AssertionError(
            "Cell 47 checkpoint verification failed "
            f"for {request['candidate_id']}: "
            f"{verified_errors}"
        )

    return (
        request["candidate_id"],
        verified_checkpoint["result"],
    )


if cell47_missing_requests:

    completed_count = 0

    with ThreadPoolExecutor(
        max_workers=CELL47_MAX_WORKERS
    ) as executor:

        future_to_request = {
            executor.submit(
                cell47_execute_and_checkpoint,
                request,
            ): request
            for request
            in cell47_missing_requests
        }

        for future in as_completed(
            future_to_request
        ):

            request = (
                future_to_request[
                    future
                ]
            )

            candidate_id = (
                request["candidate_id"]
            )

            try:

                (
                    resolved_candidate_id,
                    result,
                ) = future.result()

            except Exception as exc:

                raise RuntimeError(
                    "Cell 47 worker/checkpoint failure for "
                    f"{candidate_id}: "
                    f"{type(exc).__name__}: {exc}"
                ) from exc

            assert (
                resolved_candidate_id
                == candidate_id
            )

            completed_count += 1
            cell47_api_candidates_this_execution += 1

            cell47_api_attempts_this_execution += int(
                result.get(
                    "api_attempts",
                    0,
                )
                or 0
            )

            cell47_checkpoints_created_this_execution += 1

            print(
                f"[{completed_count:02d}/"
                f"{len(cell47_missing_requests):02d}]",
                candidate_id,
                "| structural=",
                result.get(
                    "structural_status"
                ),
                "| finish=",
                result.get(
                    "finish_reason"
                ),
                flush=True,
            )


# =========================================================
# FINAL CHECKPOINT RESCAN
# =========================================================

cell47_recovery_results = []

cell47_valid_checkpoint_count = 0
cell47_missing_checkpoint_count = 0
cell47_invalid_checkpoint_count = 0


for request in cell47_requests:

    checkpoint, status, errors = (
        cell47_load_checkpoint(
            request
        )
    )

    if status == "VALID":

        cell47_valid_checkpoint_count += 1

        cell47_recovery_results.append(
            checkpoint["result"]
        )

    elif status == "MISSING":

        cell47_missing_checkpoint_count += 1

    else:

        cell47_invalid_checkpoint_count += 1


# =========================================================
# TWO-CASE RESULT
# =========================================================

cell47_structural_distribution = Counter(
    result.get(
        "structural_status"
    )
    for result
    in cell47_recovery_results
)

cell47_finish_reason_distribution = Counter(
    result.get(
        "finish_reason"
    )
    for result
    in cell47_recovery_results
)

cell47_execution_status_distribution = Counter(
    result.get(
        "execution_status"
    )
    for result
    in cell47_recovery_results
)


cell47_structural_pass_count = (
    cell47_structural_distribution.get(
        "PASS",
        0,
    )
)

cell47_structural_fail_count = (
    cell47_structural_distribution.get(
        "FAIL",
        0,
    )
)

cell47_api_failure_count = (
    cell47_execution_status_distribution.get(
        "API_FAILED",
        0,
    )
)


# =========================================================
# COST
# =========================================================

cell47_known_costs = [
    result.get(
        "estimated_cost_toman"
    )
    for result
    in cell47_recovery_results
    if isinstance(
        result.get(
            "estimated_cost_toman"
        ),
        (int, float),
    )
]

cell47_total_estimated_cost_toman = (
    sum(cell47_known_costs)
    if cell47_known_costs
    else None
)


# =========================================================
# RECONSTRUCT FINAL 3813
# =========================================================

cell47_recovery_by_candidate_id = {
    result["candidate_id"]: result
    for result
    in cell47_recovery_results
}

assert len(
    cell47_recovery_by_candidate_id
) == len(
    cell47_recovery_results
)


cell47_final_generation_results = []


for previous_result in (
    cell46_final_generation_results
):

    candidate_id = (
        previous_result[
            "candidate_id"
        ]
    )

    if (
        candidate_id
        in cell47_recovery_by_candidate_id
    ):

        cell47_final_generation_results.append(
            cell47_recovery_by_candidate_id[
                candidate_id
            ]
        )

    else:

        cell47_final_generation_results.append(
            previous_result
        )


assert len(
    cell47_final_generation_results
) == 3813

assert len({
    result["candidate_id"]
    for result
    in cell47_final_generation_results
}) == 3813


# =========================================================
# FINAL 3813 STATE
# =========================================================

cell47_final_structural_distribution = Counter(
    result.get(
        "structural_status"
    )
    for result
    in cell47_final_generation_results
)

cell47_final_execution_status_distribution = Counter(
    result.get(
        "execution_status"
    )
    for result
    in cell47_final_generation_results
)

cell47_final_finish_reason_distribution = Counter(
    result.get(
        "finish_reason"
    )
    for result
    in cell47_final_generation_results
)


cell47_final_structural_pass_count = (
    cell47_final_structural_distribution.get(
        "PASS",
        0,
    )
)

cell47_final_structural_fail_count = (
    cell47_final_structural_distribution.get(
        "FAIL",
        0,
    )
)

cell47_final_api_failure_count = (
    cell47_final_execution_status_distribution.get(
        "API_FAILED",
        0,
    )
)


# =========================================================
# RUNNER HEALTH
# =========================================================

cell47_runner_checks = {
    "exactly_2_recovery_targets": (
        len(cell47_requests) == 2
    ),

    "2_results_reconstructed": (
        len(cell47_recovery_results) == 2
    ),

    "2_valid_checkpoints": (
        cell47_valid_checkpoint_count == 2
    ),

    "zero_missing_checkpoints": (
        cell47_missing_checkpoint_count == 0
    ),

    "zero_invalid_checkpoints": (
        cell47_invalid_checkpoint_count == 0
    ),

    "final_population_3813": (
        len(
            cell47_final_generation_results
        )
        == 3813
    ),

    "final_unique_candidates_3813": (
        len({
            result["candidate_id"]
            for result
            in cell47_final_generation_results
        })
        == 3813
    ),

    "3811_previous_passes_preserved": (
        sum(
            1
            for result
            in cell46_final_generation_results
            if (
                result.get(
                    "structural_status"
                )
                == "PASS"
            )
        )
        == 3811
    ),
}


cell47_runner_status = (
    "PASS"
    if all(
        cell47_runner_checks.values()
    )
    else "FAIL"
)


cell47_recovery_status = (
    "PASS"
    if (
        cell47_structural_pass_count == 2
        and
        cell47_api_failure_count == 0
    )
    else "FAIL"
)


cell47_final_generation_status = (
    "PASS"
    if (
        cell47_final_structural_pass_count
        == 3813
        and
        cell47_final_structural_fail_count
        == 0
        and
        cell47_final_api_failure_count
        == 0
    )
    else "FAIL"
)


if cell47_runner_status != "PASS":

    cell47_next_readiness = (
        "BLOCKED_CELL47_RUNNER_REPAIR_REQUIRED"
    )

elif cell47_api_failure_count > 0:

    cell47_next_readiness = (
        "READY_FOR_CELL47_API_FAILURE_RECOVERY"
    )

elif cell47_structural_fail_count > 0:

    cell47_next_readiness = (
        "READY_FOR_CELL47_STRUCTURAL_FAILURE_ANALYSIS"
    )

elif cell47_final_generation_status == "PASS":

    cell47_next_readiness = (
        "READY_FOR_FULL_QA_SEMANTIC_VALIDATION_V4"
    )

else:

    cell47_next_readiness = (
        "READY_FOR_FINAL_GENERATION_STATE_ANALYSIS"
    )


# =========================================================
# OUTPUT
# =========================================================

print()

print(
    "=== GOLDEN V2 FINAL TWO-CASE "
    "GENERATION RECOVERY RESULT ==="
)

print()

print(
    "Recovery candidates:",
    cell47_candidate_ids,
)

print(
    "Recovery structural distribution:",
    dict(
        cell47_structural_distribution
    ),
)

print(
    "Recovery finish reasons:",
    dict(
        cell47_finish_reason_distribution
    ),
)

print(
    "Recovery API failures:",
    cell47_api_failure_count,
)

print(
    "API candidates this execution:",
    cell47_api_candidates_this_execution,
)

print(
    "API attempts this execution:",
    cell47_api_attempts_this_execution,
)

print(
    "Estimated recovery cost represented "
    "by checkpoints (toman):",
    cell47_total_estimated_cost_toman,
)

print()

print(
    "=== FINAL 3813 GENERATION STATE ==="
)

print(
    "Population:",
    len(
        cell47_final_generation_results
    ),
)

print(
    "Structural distribution:",
    dict(
        cell47_final_structural_distribution
    ),
)

print(
    "Structural PASS:",
    cell47_final_structural_pass_count,
    "/ 3813",
)

print(
    "Structural FAIL:",
    cell47_final_structural_fail_count,
    "/ 3813",
)

print(
    "API failures:",
    cell47_final_api_failure_count,
)

print(
    "Finish reason distribution:",
    dict(
        cell47_final_finish_reason_distribution
    ),
)

print()

print("Runner checks:")

for (
    check_name,
    check_value,
) in cell47_runner_checks.items():

    print(
        " -",
        check_name,
        ":",
        (
            "PASS"
            if check_value
            else "FAIL"
        ),
    )

print()

print(
    "Runner health:",
    cell47_runner_status,
)

print(
    "Two-case recovery result:",
    cell47_recovery_status,
)

print(
    "Final Full Generation result:",
    cell47_final_generation_status,
)

print(
    "Next readiness:",
    cell47_next_readiness,
)

print(
    "CELL 47 CHECKPOINTED RUNNER:",
    cell47_runner_status,
)


assert cell47_runner_status == "PASS", (
    "Cell 47 checkpointed runner health failed."
)
````

<h2>مرحله ۴۸ — اعتبارسنجی معنایی کامل ۳۸۱۳ Question/Answer نهایی</h2>
<p>
پس از تکمیل موفق تولید ساختاری هر ۳۸۱۳ Question/Answer، در این مرحله تمام خروجی‌های نهایی در برابر Claim و Evidence متناظر اعتبارسنجی معنایی می‌شوند.
این اعتبارسنجی فقط کیفیت و وفاداری Question/Answer تولیدشده را بررسی می‌کند و اعتبار Claim یا Evidence را از نو ارزیابی نمی‌کند؛ زیرا این لایه در مراحل پیشین Claim Registry به‌صورت کامل اعتبارسنجی شده است.
برای هر Candidate هشت بُعد شامل هدف‌گیری صحیح Claim، بازیابی معنای Claim در Answer، وفاداری Scope، حفظ Modality و Quantifier، مستقل‌بودن Question، نبود Neighbor Fact Drift، نبود External Knowledge و هم‌راستایی Answer با Question بررسی می‌شود.
تمام Judgmentها دارای Fingerprint و Checkpoint مستقل هستند و اجرای مجدد فقط موارد فاقد Checkpoint معتبر را به مدل ارسال خواهد کرد.
</p>
</div>

````python
# Code Cell 48 — Full QA Semantic Validation v4
#
# FULL CORRECTED VERSION
#
# IMPORTANT:
# - Existing exact checkpoints remain authoritative.
# - Existing request fingerprints are preserved.
# - Only missing exact checkpoints may call the API.
# - Empty/invalid gateway responses no longer crash the runner.
# - Individual worker failures no longer kill the entire run.
# - API requests have an explicit timeout.
# - 16 workers / 0.6 s request-start spacing are preserved.

from collections import Counter
from concurrent.futures import ThreadPoolExecutor, as_completed
from pathlib import Path
import hashlib
import json
import os
import re
import tempfile
import threading
import time


# =========================================================
# CONFIG
# =========================================================

CELL48_RUNNER_VERSION = (
    "golden_v2_full_qa_semantic_validator_v1"
)

CELL48_VALIDATION_PROTOCOL_VERSION = (
    "golden_v2_qa_semantic_validation_v1"
)

CELL48_OUTPUT_SCHEMA_VERSION = (
    "golden_v2_qa_semantic_validation_output_v1"
)

CELL48_MAX_TOKENS = 1536
CELL48_TEMPERATURE = 0.0
CELL48_TOP_P = 1.0

CELL48_MAX_ATTEMPTS = 2
CELL48_RETRY_SECONDS = 2.0

# Gateway limit = 120 requests/minute.
CELL48_MAX_WORKERS = 16
CELL48_REQUEST_START_SPACING_SECONDS = 0.6

# Prevent a network request from holding a worker indefinitely.
CELL48_REQUEST_TIMEOUT_SECONDS = 180.0

# Intentional paid Full Semantic Validation.
# MUST become False before final Restart -> Run All.
CELL48_ALLOW_API_FOR_MISSING = False


# =========================================================
# REQUIRED STATE
# =========================================================

required_cell48_names = [
    "project_root",
    "semantic_review_client",
    "semantic_review_model",
    "cell25_candidate_skeletons",
    "cell47_final_generation_results",
    "CELL37_SYSTEM_PROMPT",
    "CELL37_BOOLEAN_FIELDS",
    "cell37_validate_output",
]

missing_cell48_names = [
    name
    for name in required_cell48_names
    if name not in globals()
]

assert not missing_cell48_names, (
    "Missing required Cell 48 state: "
    + ", ".join(missing_cell48_names)
)

assert len(cell25_candidate_skeletons) == 3813
assert len(cell47_final_generation_results) == 3813

assert all(
    result.get("structural_status") == "PASS"
    for result in cell47_final_generation_results
), (
    "Cell 48 requires all 3813 generated QA outputs "
    "to be structurally valid."
)

assert len({
    record["candidate_id"]
    for record in cell25_candidate_skeletons
}) == 3813

assert len({
    result["candidate_id"]
    for result in cell47_final_generation_results
}) == 3813


# =========================================================
# EXACT LOOKUPS
# =========================================================

cell48_skeleton_by_candidate_id = {
    record["candidate_id"]: record
    for record in cell25_candidate_skeletons
}

cell48_generation_by_candidate_id = {
    result["candidate_id"]: result
    for result in cell47_final_generation_results
}

assert (
    set(cell48_skeleton_by_candidate_id)
    ==
    set(cell48_generation_by_candidate_id)
)

cell48_candidate_ids = [
    record["candidate_id"]
    for record in cell25_candidate_skeletons
]


# =========================================================
# VERIFY FINAL GENERATED QA PAYLOADS
# =========================================================

cell48_generation_payload_errors = []

for candidate_id in cell48_candidate_ids:

    generation_result = (
        cell48_generation_by_candidate_id[
            candidate_id
        ]
    )

    generated_qa = generation_result.get(
        "parsed_json"
    )

    if not isinstance(generated_qa, dict):

        cell48_generation_payload_errors.append({
            "candidate_id": candidate_id,
            "error": "PARSED_JSON_NOT_OBJECT",
        })

        continue

    for required_field in [
        "question",
        "expected_answer",
        "question_intent",
    ]:

        value = generated_qa.get(
            required_field
        )

        if (
            not isinstance(value, str)
            or not value.strip()
        ):

            cell48_generation_payload_errors.append({
                "candidate_id": candidate_id,
                "error": (
                    "INVALID_GENERATED_FIELD:"
                    + required_field
                ),
            })


assert not cell48_generation_payload_errors, (
    "Invalid final generation payloads detected: "
    f"{cell48_generation_payload_errors[:10]}"
)


# =========================================================
# HASH UTILITIES
# =========================================================

def cell48_canonical_json(value):

    return json.dumps(
        value,
        ensure_ascii=False,
        sort_keys=True,
        separators=(",", ":"),
    )


def cell48_sha256_text(text):

    return hashlib.sha256(
        text.encode("utf-8")
    ).hexdigest()


# =========================================================
# VALIDATION PACKET
# =========================================================

def cell48_build_validation_packet(
    candidate_id,
):

    skeleton = (
        cell48_skeleton_by_candidate_id[
            candidate_id
        ]
    )

    generation_result = (
        cell48_generation_by_candidate_id[
            candidate_id
        ]
    )

    generated_qa = (
        generation_result[
            "parsed_json"
        ]
    )

    evidence = []

    for evidence_index, (
        evidence_id,
        evidence_quote,
    ) in enumerate(
        zip(
            skeleton["evidence_ids"],
            skeleton["evidence_quotes"],
        ),
        start=1,
    ):

        evidence.append({
            "index": evidence_index,
            "evidence_id": evidence_id,
            "quote": evidence_quote,
        })

    packet = {
        "candidate_id": candidate_id,

        "claim": {
            "claim_id": (
                skeleton["claim_id"]
            ),

            "claim_text_fa": (
                skeleton["claim_text_fa"]
            ),

            "claim_type": (
                skeleton.get(
                    "claim_type"
                )
            ),

            "claim_types": (
                skeleton.get(
                    "claim_types",
                    [],
                )
            ),

            "authority_class": (
                skeleton.get(
                    "authority_class"
                )
            ),
        },

        "evidence": evidence,

        "generated_qa": {
            "question": (
                generated_qa[
                    "question"
                ]
            ),

            "expected_answer": (
                generated_qa[
                    "expected_answer"
                ]
            ),

            "question_intent": (
                generated_qa[
                    "question_intent"
                ]
            ),
        },
    }

    return packet


# =========================================================
# USER PROMPT
# =========================================================

def cell48_build_user_prompt(
    candidate_id,
):

    packet = (
        cell48_build_validation_packet(
            candidate_id
        )
    )

    return (
        "Validate ONLY the generated Question and Expected Answer "
        "against the supplied prevalidated Claim and Evidence.\n\n"

        "The Claim and Evidence have already passed upstream semantic "
        "validation. Do NOT decide whether the Claim itself should be "
        "accepted or rejected. Do NOT perform a new Claim-Evidence "
        "eligibility review.\n\n"

        "Evaluate exactly these QA dimensions:\n"
        "1. question_targets_claim\n"
        "2. answer_recovers_claim\n"
        "3. scope_fidelity\n"
        "4. modality_quantifier_fidelity\n"
        "5. standalone_question\n"
        "6. no_neighbor_fact_drift\n"
        "7. no_external_knowledge\n"
        "8. answer_question_alignment\n\n"

        "Important fidelity rules:\n"
        "- The Question must ask for information actually asserted "
        "by the Claim.\n"
        "- The Expected Answer must correctly answer that Question "
        "and recover the relevant meaning of the Claim.\n"
        "- Preserve meaning-bearing negation, modality, attribution, "
        "conditions, exceptions, scope, quantities, percentages, "
        "comparisons and quantifiers such as only/تنها/فقط when "
        "they are relevant to the generated QA target.\n"
        "- Do not permit WHY/HOW/cause/reason/mechanism claims unless "
        "the supplied Claim actually asserts them.\n"
        "- Do not allow facts imported from neighboring Evidence text "
        "when those facts are not part of the target Claim.\n"
        "- Do not require external knowledge.\n"
        "- A narrower Question/Answer may be valid only when it "
        "faithfully targets a self-contained asserted part of the "
        "Claim without changing its meaning or introducing drift.\n\n"

        "Return only the JSON object required by the validation "
        "system prompt.\n\n"

        "VALIDATION PACKET:\n"
        + json.dumps(
            packet,
            ensure_ascii=False,
            sort_keys=True,
            indent=2,
        )
    )


# =========================================================
# EXACT REQUEST CONTRACT
# =========================================================

def cell48_build_request(
    candidate_id,
):

    packet = (
        cell48_build_validation_packet(
            candidate_id
        )
    )

    user_prompt = (
        cell48_build_user_prompt(
            candidate_id
        )
    )

    generation_result = (
        cell48_generation_by_candidate_id[
            candidate_id
        ]
    )

    generated_qa = (
        generation_result[
            "parsed_json"
        ]
    )

    generated_qa_sha256 = (
        cell48_sha256_text(
            cell48_canonical_json(
                generated_qa
            )
        )
    )

    validation_packet_sha256 = (
        cell48_sha256_text(
            cell48_canonical_json(
                packet
            )
        )
    )

    contract = {
        "runner_version": (
            CELL48_RUNNER_VERSION
        ),

        "validation_protocol_version": (
            CELL48_VALIDATION_PROTOCOL_VERSION
        ),

        "output_schema_version": (
            CELL48_OUTPUT_SCHEMA_VERSION
        ),

        "candidate_id": candidate_id,

        "generated_qa_sha256": (
            generated_qa_sha256
        ),

        "validation_packet_sha256": (
            validation_packet_sha256
        ),

        "model": semantic_review_model,

        "system_prompt": (
            CELL37_SYSTEM_PROMPT
        ),

        "user_prompt": user_prompt,

        "temperature": (
            CELL48_TEMPERATURE
        ),

        "top_p": CELL48_TOP_P,

        "max_tokens": (
            CELL48_MAX_TOKENS
        ),

        "enable_thinking": False,
    }

    request_fingerprint = (
        cell48_sha256_text(
            cell48_canonical_json(
                contract
            )
        )
    )

    return {
        **contract,
        "request_fingerprint": (
            request_fingerprint
        ),
    }


cell48_requests = [
    cell48_build_request(
        candidate_id
    )
    for candidate_id
    in cell48_candidate_ids
]

assert len(cell48_requests) == 3813

assert len({
    request["candidate_id"]
    for request in cell48_requests
}) == 3813

assert len({
    request["request_fingerprint"]
    for request in cell48_requests
}) == 3813


cell48_request_by_candidate_id = {
    request["candidate_id"]: request
    for request in cell48_requests
}


# =========================================================
# CHECKPOINT DIRECTORY
# =========================================================

cell48_checkpoint_dir = (
    Path(project_root)
    / "artifacts"
    / "phase2"
    / "golden_dataset_v2"
    / "work"
    / "full_qa_semantic_validation_v4"
)

cell48_checkpoint_dir.mkdir(
    parents=True,
    exist_ok=True,
)


def cell48_checkpoint_path(
    candidate_id,
):

    safe_id = re.sub(
        r"[^A-Za-z0-9_.-]+",
        "_",
        candidate_id,
    )

    return (
        cell48_checkpoint_dir
        / f"{safe_id}.json"
    )


# =========================================================
# ATOMIC WRITE
# =========================================================

def cell48_atomic_write_json(
    path,
    payload,
):

    path = Path(path)

    content = (
        json.dumps(
            payload,
            ensure_ascii=False,
            sort_keys=True,
            indent=2,
        )
        + "\n"
    )

    fd, temp_name = tempfile.mkstemp(
        prefix=path.name + ".",
        suffix=".tmp",
        dir=str(path.parent),
    )

    try:

        with os.fdopen(
            fd,
            "w",
            encoding="utf-8",
        ) as handle:

            handle.write(content)
            handle.flush()
            os.fsync(
                handle.fileno()
            )

        os.replace(
            temp_name,
            path,
        )

    except Exception:

        try:
            os.unlink(
                temp_name
            )
        except FileNotFoundError:
            pass

        raise


# =========================================================
# CHECKPOINT CONTRACT
# =========================================================

def cell48_build_checkpoint(
    request,
    result,
):

    checkpoint = {
        "checkpoint_version": (
            CELL48_RUNNER_VERSION
        ),

        "candidate_id": (
            request["candidate_id"]
        ),

        "request_fingerprint": (
            request["request_fingerprint"]
        ),

        "generated_qa_sha256": (
            request["generated_qa_sha256"]
        ),

        "validation_packet_sha256": (
            request[
                "validation_packet_sha256"
            ]
        ),

        "validation_protocol_version": (
            CELL48_VALIDATION_PROTOCOL_VERSION
        ),

        "output_schema_version": (
            CELL48_OUTPUT_SCHEMA_VERSION
        ),

        "model": semantic_review_model,

        "temperature": (
            CELL48_TEMPERATURE
        ),

        "top_p": CELL48_TOP_P,

        "max_tokens": (
            CELL48_MAX_TOKENS
        ),

        "result": result,
    }

    checkpoint[
        "checkpoint_payload_sha256"
    ] = cell48_sha256_text(
        cell48_canonical_json(
            checkpoint
        )
    )

    return checkpoint


def cell48_validate_checkpoint(
    checkpoint,
    request,
):

    errors = []

    if not isinstance(
        checkpoint,
        dict,
    ):

        return [
            "CHECKPOINT_NOT_OBJECT"
        ]

    stored_sha = checkpoint.get(
        "checkpoint_payload_sha256"
    )

    payload_without_sha = {
        key: value
        for key, value
        in checkpoint.items()
        if key
        != "checkpoint_payload_sha256"
    }

    recalculated_sha = (
        cell48_sha256_text(
            cell48_canonical_json(
                payload_without_sha
            )
        )
    )

    if stored_sha != recalculated_sha:

        errors.append(
            "CHECKPOINT_HASH_MISMATCH"
        )

    exact_expectations = {
        "checkpoint_version": (
            CELL48_RUNNER_VERSION
        ),

        "candidate_id": (
            request["candidate_id"]
        ),

        "request_fingerprint": (
            request["request_fingerprint"]
        ),

        "generated_qa_sha256": (
            request["generated_qa_sha256"]
        ),

        "validation_packet_sha256": (
            request[
                "validation_packet_sha256"
            ]
        ),

        "validation_protocol_version": (
            CELL48_VALIDATION_PROTOCOL_VERSION
        ),

        "output_schema_version": (
            CELL48_OUTPUT_SCHEMA_VERSION
        ),

        "model": semantic_review_model,

        "temperature": (
            CELL48_TEMPERATURE
        ),

        "top_p": CELL48_TOP_P,

        "max_tokens": (
            CELL48_MAX_TOKENS
        ),
    }

    for (
        field_name,
        expected_value,
    ) in exact_expectations.items():

        if (
            checkpoint.get(
                field_name
            )
            != expected_value
        ):

            errors.append(
                f"{field_name.upper()}_MISMATCH"
            )

    result = checkpoint.get(
        "result"
    )

    if not isinstance(
        result,
        dict,
    ):

        errors.append(
            "CHECKPOINT_RESULT_INVALID"
        )

    elif (
        result.get(
            "candidate_id"
        )
        != request["candidate_id"]
    ):

        errors.append(
            "CHECKPOINT_RESULT_CANDIDATE_MISMATCH"
        )

    return sorted(
        set(errors)
    )


def cell48_load_checkpoint(
    request,
):

    path = cell48_checkpoint_path(
        request["candidate_id"]
    )

    if not path.exists():

        return (
            None,
            "MISSING",
            [],
        )

    try:

        with path.open(
            "r",
            encoding="utf-8",
        ) as handle:

            checkpoint = json.load(
                handle
            )

    except Exception as exc:

        return (
            None,
            "UNREADABLE",
            [
                f"{type(exc).__name__}: {exc}"
            ],
        )

    errors = (
        cell48_validate_checkpoint(
            checkpoint,
            request,
        )
    )

    if errors:

        return (
            checkpoint,
            "INVALID",
            errors,
        )

    return (
        checkpoint,
        "VALID",
        [],
    )


# =========================================================
# MODEL OUTPUT CLEANING
# =========================================================

def cell48_clean_model_json(
    content,
):

    if not isinstance(
        content,
        str,
    ):
        return ""

    clean = content.strip()

    if clean.startswith("```"):

        clean = re.sub(
            r"^```(?:json)?\s*",
            "",
            clean,
            flags=re.IGNORECASE,
        )

        clean = re.sub(
            r"\s*```$",
            "",
            clean,
        )

    return clean.strip()


def cell48_is_rate_limit_exception(
    exc,
):

    error_text = str(
        exc
    ).lower()

    error_type = type(
        exc
    ).__name__.lower()

    return (
        "ratelimit" in error_type
        or "rate_limit" in error_type
        or "429" in error_text
        or "too many requests" in error_text
    )


# =========================================================
# GLOBAL REQUEST-START SPACING
# =========================================================

cell48_request_start_lock = (
    threading.Lock()
)

cell48_last_request_start = [
    0.0
]


def cell48_wait_for_request_slot():

    with cell48_request_start_lock:

        now = time.monotonic()

        elapsed = (
            now
            - cell48_last_request_start[0]
        )

        wait_seconds = max(
            0.0,
            (
                CELL48_REQUEST_START_SPACING_SECONDS
                - elapsed
            ),
        )

        if wait_seconds > 0:

            time.sleep(
                wait_seconds
            )

        cell48_last_request_start[0] = (
            time.monotonic()
        )


# =========================================================
# COST
# =========================================================

cell48_has_cost_rates = (
    "INPUT_TOMAN_PER_TOKEN"
    in globals()
    and
    "OUTPUT_TOMAN_PER_TOKEN"
    in globals()
)


def cell48_estimated_cost_toman(
    input_tokens,
    output_tokens,
):

    if not cell48_has_cost_rates:
        return None

    return (
        input_tokens
        * INPUT_TOMAN_PER_TOKEN
        +
        output_tokens
        * OUTPUT_TOMAN_PER_TOKEN
    )


# =========================================================
# API FAILURE RESULT
# =========================================================

def cell48_build_api_failure_result(
    request,
    api_attempts,
    attempt_history,
    error_code,
    elapsed_seconds=None,
):

    return {
        "candidate_id": (
            request["candidate_id"]
        ),

        "request_fingerprint": (
            request["request_fingerprint"]
        ),

        "validation_protocol_version": (
            CELL48_VALIDATION_PROTOCOL_VERSION
        ),

        "execution_status": (
            "API_FAILED"
        ),

        "api_attempts": (
            api_attempts
        ),

        "attempt_history": (
            attempt_history
        ),

        "finish_reason": None,

        "elapsed_seconds": (
            round(
                elapsed_seconds,
                4,
            )
            if isinstance(
                elapsed_seconds,
                (int, float),
            )
            else None
        ),

        "usage": {
            "input_tokens": 0,
            "output_tokens": 0,
            "reasoning_tokens": 0,
            "total_tokens": 0,
        },

        "estimated_cost_toman": None,

        "raw_content": None,

        "parsed_json": None,

        "parse_error": None,

        "structural_errors": [
            error_code
        ],

        "structural_status": (
            "FAIL"
        ),

        "semantic_verdict": None,

        "validation": None,
    }


# =========================================================
# API EXECUTION
# =========================================================

def cell48_execute_request(
    request,
):

    attempt_history = []
    api_attempts = 0

    for attempt_number in range(
        1,
        CELL48_MAX_ATTEMPTS + 1,
    ):

        cell48_wait_for_request_slot()

        started_at = time.time()
        api_attempts += 1

        try:

            response = (
                semantic_review_client
                .chat
                .completions
                .create(
                    model=(
                        semantic_review_model
                    ),

                    messages=[
                        {
                            "role": "system",
                            "content": (
                                CELL37_SYSTEM_PROMPT
                            ),
                        },
                        {
                            "role": "user",
                            "content": (
                                request[
                                    "user_prompt"
                                ]
                            ),
                        },
                    ],

                    temperature=(
                        CELL48_TEMPERATURE
                    ),

                    top_p=(
                        CELL48_TOP_P
                    ),

                    max_tokens=(
                        CELL48_MAX_TOKENS
                    ),

                    timeout=(
                        CELL48_REQUEST_TIMEOUT_SECONDS
                    ),

                    extra_body={
                        "enable_thinking": False,
                    },
                )
            )

            if response is None:

                raise RuntimeError(
                    "EMPTY_API_RESPONSE"
                )

            response_choices = getattr(
                response,
                "choices",
                None,
            )

            if (
                not isinstance(
                    response_choices,
                    (list, tuple),
                )
                or
                len(response_choices) == 0
            ):

                raise RuntimeError(
                    "API_RESPONSE_CHOICES_MISSING_OR_EMPTY"
                )

            choice = response_choices[0]

            if choice is None:

                raise RuntimeError(
                    "API_RESPONSE_FIRST_CHOICE_IS_NONE"
                )

            message = getattr(
                choice,
                "message",
                None,
            )

            if message is None:

                raise RuntimeError(
                    "API_RESPONSE_MESSAGE_MISSING"
                )

        except Exception as exc:

            elapsed_seconds = (
                time.time()
                - started_at
            )

            attempt_history.append({
                "attempt": (
                    attempt_number
                ),

                "status": (
                    "API_ERROR"
                ),

                "error": (
                    f"{type(exc).__name__}: {exc}"
                ),
            })

            if (
                attempt_number
                < CELL48_MAX_ATTEMPTS
            ):

                wait_seconds = (
                    70.0
                    if cell48_is_rate_limit_exception(
                        exc
                    )
                    else (
                        CELL48_RETRY_SECONDS
                        * attempt_number
                    )
                )

                time.sleep(
                    wait_seconds
                )

                continue

            return (
                cell48_build_api_failure_result(
                    request=request,
                    api_attempts=api_attempts,
                    attempt_history=attempt_history,
                    error_code="API_FAILED",
                    elapsed_seconds=elapsed_seconds,
                )
            )

        elapsed_seconds = (
            time.time()
            - started_at
        )

        finish_reason = getattr(
            choice,
            "finish_reason",
            None,
        )

        raw_content = getattr(
            message,
            "content",
            None,
        )

        usage = getattr(
            response,
            "usage",
            None,
        )

        input_tokens = int(
            getattr(
                usage,
                "prompt_tokens",
                0,
            )
            or 0
        )

        output_tokens = int(
            getattr(
                usage,
                "completion_tokens",
                0,
            )
            or 0
        )

        total_tokens = int(
            getattr(
                usage,
                "total_tokens",
                0,
            )
            or (
                input_tokens
                + output_tokens
            )
        )

        reasoning_tokens = 0

        completion_details = getattr(
            usage,
            "completion_tokens_details",
            None,
        )

        if completion_details is not None:

            reasoning_tokens = int(
                getattr(
                    completion_details,
                    "reasoning_tokens",
                    0,
                )
                or 0
            )

        clean_content = (
            cell48_clean_model_json(
                raw_content
            )
        )

        parsed_json = None
        parse_error = None

        try:

            parsed_json = json.loads(
                clean_content
            )

        except Exception as exc:

            parse_error = (
                f"{type(exc).__name__}: {exc}"
            )

        if parsed_json is None:

            structural_errors = [
                "JSON_PARSE_FAILED"
            ]

        else:

            structural_errors = (
                cell37_validate_output(
                    parsed_json
                )
            )

        structural_status = (
            "PASS"
            if not structural_errors
            else "FAIL"
        )

        semantic_verdict = None

        if (
            structural_status
            == "PASS"
        ):

            semantic_verdict = (
                parsed_json.get(
                    "verdict"
                )
            )

        return {
            "candidate_id": (
                request["candidate_id"]
            ),

            "request_fingerprint": (
                request[
                    "request_fingerprint"
                ]
            ),

            "validation_protocol_version": (
                CELL48_VALIDATION_PROTOCOL_VERSION
            ),

            "execution_status": (
                "COMPLETED"
            ),

            "api_attempts": (
                api_attempts
            ),

            "finish_reason": (
                finish_reason
            ),

            "elapsed_seconds": round(
                elapsed_seconds,
                4,
            ),

            "usage": {
                "input_tokens": (
                    input_tokens
                ),

                "output_tokens": (
                    output_tokens
                ),

                "reasoning_tokens": (
                    reasoning_tokens
                ),

                "total_tokens": (
                    total_tokens
                ),
            },

            "estimated_cost_toman": (
                cell48_estimated_cost_toman(
                    input_tokens,
                    output_tokens,
                )
            ),

            "raw_content": (
                raw_content
            ),

            "parsed_json": (
                parsed_json
            ),

            "parse_error": (
                parse_error
            ),

            "structural_errors": (
                structural_errors
            ),

            "structural_status": (
                structural_status
            ),

            "semantic_verdict": (
                semantic_verdict
            ),

            "validation": (
                parsed_json
                if structural_status
                == "PASS"
                else None
            ),

            "attempt_history": (
                attempt_history
            ),
        }

    return (
        cell48_build_api_failure_result(
            request=request,
            api_attempts=api_attempts,
            attempt_history=attempt_history,
            error_code=(
                "UNEXPECTED_EXECUTION_PATH"
            ),
        )
    )


# =========================================================
# PREFLIGHT SCAN
# =========================================================

cell48_valid_before_api = {}
cell48_missing_requests = []
cell48_invalid_before_api = []


for request in cell48_requests:

    (
        checkpoint,
        status,
        errors,
    ) = cell48_load_checkpoint(
        request
    )

    candidate_id = (
        request["candidate_id"]
    )

    if status == "VALID":

        cell48_valid_before_api[
            candidate_id
        ] = checkpoint

    elif status == "MISSING":

        cell48_missing_requests.append(
            request
        )

    else:

        cell48_invalid_before_api.append({
            "candidate_id": candidate_id,
            "status": status,
            "errors": errors,
        })


assert not cell48_invalid_before_api, (
    "Invalid/unreadable Cell 48 checkpoints detected. "
    "Refusing automatic paid replacement. "
    f"Examples: {cell48_invalid_before_api[:5]}"
)


# =========================================================
# PREFLIGHT OUTPUT
# =========================================================

print(
    "=== GOLDEN V2 FULL QA SEMANTIC "
    "VALIDATION PREFLIGHT ==="
)

print()

print(
    "Validation population:",
    len(cell48_requests),
)

print(
    "Generated QA structural PASS:",
    sum(
        1
        for result
        in cell47_final_generation_results
        if result.get(
            "structural_status"
        )
        == "PASS"
    ),
)

print(
    "Exact valid semantic checkpoints:",
    len(
        cell48_valid_before_api
    ),
)

print(
    "Missing semantic checkpoints:",
    len(
        cell48_missing_requests
    ),
)

print(
    "Model:",
    semantic_review_model,
)

print(
    "Validation protocol:",
    CELL48_VALIDATION_PROTOCOL_VERSION,
)

print(
    "Max tokens:",
    CELL48_MAX_TOKENS,
)

print(
    "Parallel workers:",
    CELL48_MAX_WORKERS,
)

print(
    "Minimum seconds between request starts:",
    CELL48_REQUEST_START_SPACING_SECONDS,
)

print(
    "Request timeout seconds:",
    CELL48_REQUEST_TIMEOUT_SECONDS,
)

print(
    "Target request starts/minute:",
    round(
        60.0
        / CELL48_REQUEST_START_SPACING_SECONDS,
        2,
    ),
)

print(
    "Gateway configured limit/minute:",
    120,
)

print(
    "API allowed for missing exact checkpoints:",
    CELL48_ALLOW_API_FOR_MISSING,
)

print(
    "Claim-Evidence revalidation:",
    False,
)

print()


if (
    cell48_missing_requests
    and
    not CELL48_ALLOW_API_FOR_MISSING
):

    raise AssertionError(
        "Cell 48 has missing semantic-validation "
        "checkpoints but API execution is disabled."
    )


# =========================================================
# FULL SEMANTIC VALIDATION EXECUTION
# =========================================================

cell48_api_candidates_this_execution = 0
cell48_api_attempts_this_execution = 0
cell48_checkpoints_created_this_execution = 0


def cell48_execute_and_checkpoint(
    request,
):

    candidate_id = (
        request["candidate_id"]
    )

    result = (
        cell48_execute_request(
            request
        )
    )

    checkpoint = (
        cell48_build_checkpoint(
            request,
            result,
        )
    )

    cell48_atomic_write_json(
        cell48_checkpoint_path(
            candidate_id
        ),
        checkpoint,
    )

    (
        verified_checkpoint,
        verified_status,
        verified_errors,
    ) = cell48_load_checkpoint(
        request
    )

    if (
        verified_status
        != "VALID"
    ):

        raise AssertionError(
            "Cell 48 checkpoint verification "
            f"failed for {candidate_id}: "
            f"{verified_errors}"
        )

    return (
        candidate_id,
        verified_checkpoint[
            "result"
        ],
    )


if cell48_missing_requests:

    completed_count = 0

    executor = ThreadPoolExecutor(
        max_workers=(
            CELL48_MAX_WORKERS
        )
    )

    future_to_request = {}

    try:

        future_to_request = {
            executor.submit(
                cell48_execute_and_checkpoint,
                request,
            ): request

            for request
            in cell48_missing_requests
        }

        for future in as_completed(
            future_to_request
        ):

            request = (
                future_to_request[
                    future
                ]
            )

            candidate_id = (
                request[
                    "candidate_id"
                ]
            )

            try:

                (
                    resolved_candidate_id,
                    result,
                ) = future.result()

            except Exception as exc:

                completed_count += 1

                print(
                    f"[{completed_count:04d}/"
                    f"{len(cell48_missing_requests):04d}]",
                    candidate_id,
                    "| WORKER_EXCEPTION =",
                    f"{type(exc).__name__}: {exc}",
                    flush=True,
                )

                continue

            assert (
                resolved_candidate_id
                == candidate_id
            )

            completed_count += 1

            cell48_api_candidates_this_execution += 1

            cell48_api_attempts_this_execution += int(
                result.get(
                    "api_attempts",
                    0,
                )
                or 0
            )

            cell48_checkpoints_created_this_execution += 1

            if (
                completed_count <= 20
                or
                completed_count % 25 == 0
                or
                completed_count
                == len(
                    cell48_missing_requests
                )
            ):

                print(
                    f"[{completed_count:04d}/"
                    f"{len(cell48_missing_requests):04d}]",
                    candidate_id,
                    "| structural=",
                    result.get(
                        "structural_status"
                    ),
                    "| semantic=",
                    result.get(
                        "semantic_verdict"
                    ),
                    "| finish=",
                    result.get(
                        "finish_reason"
                    ),
                    flush=True,
                )

    except KeyboardInterrupt:

        print(
            "\nCell 48 interrupted by user. "
            "Completed checkpoints remain preserved.",
            flush=True,
        )

        for future in future_to_request:

            if not future.done():

                future.cancel()

        raise

    finally:

        executor.shutdown(
            wait=False,
            cancel_futures=True,
        )


# =========================================================
# FINAL DISK RESCAN
# =========================================================

cell48_full_semantic_results = []

cell48_valid_checkpoint_count = 0
cell48_missing_checkpoint_count = 0
cell48_invalid_checkpoint_count = 0


for request in cell48_requests:

    (
        checkpoint,
        status,
        errors,
    ) = cell48_load_checkpoint(
        request
    )

    if status == "VALID":

        cell48_valid_checkpoint_count += 1

        cell48_full_semantic_results.append(
            checkpoint[
                "result"
            ]
        )

    elif status == "MISSING":

        cell48_missing_checkpoint_count += 1

    else:

        cell48_invalid_checkpoint_count += 1


# =========================================================
# STRUCTURAL RESULT
# =========================================================

cell48_structural_distribution = Counter(
    result.get(
        "structural_status"
    )
    for result
    in cell48_full_semantic_results
)

cell48_execution_status_distribution = Counter(
    result.get(
        "execution_status"
    )
    for result
    in cell48_full_semantic_results
)

cell48_finish_reason_distribution = Counter(
    result.get(
        "finish_reason"
    )
    for result
    in cell48_full_semantic_results
)


cell48_structural_pass_count = (
    cell48_structural_distribution.get(
        "PASS",
        0,
    )
)

cell48_structural_fail_count = (
    cell48_structural_distribution.get(
        "FAIL",
        0,
    )
)

cell48_api_failure_count = (
    cell48_execution_status_distribution.get(
        "API_FAILED",
        0,
    )
)


# =========================================================
# SEMANTIC RESULT
# =========================================================

cell48_structurally_valid_results = [
    result
    for result
    in cell48_full_semantic_results
    if result.get(
        "structural_status"
    )
    == "PASS"
]


cell48_semantic_distribution = Counter(
    result.get(
        "semantic_verdict"
    )
    for result
    in cell48_structurally_valid_results
)


cell48_semantic_pass_results = [
    result
    for result
    in cell48_structurally_valid_results
    if result.get(
        "semantic_verdict"
    )
    == "PASS"
]


cell48_semantic_fail_results = [
    result
    for result
    in cell48_structurally_valid_results
    if result.get(
        "semantic_verdict"
    )
    == "FAIL"
]


cell48_semantic_pass_count = len(
    cell48_semantic_pass_results
)

cell48_semantic_fail_count = len(
    cell48_semantic_fail_results
)


# =========================================================
# DIMENSION FAILURE DISTRIBUTION
# =========================================================

cell48_dimension_failure_distribution = Counter()

cell48_failure_reason_distribution = Counter()


for result in cell48_semantic_fail_results:

    validation = (
        result.get(
            "validation"
        )
        or {}
    )

    for field_name in (
        CELL37_BOOLEAN_FIELDS
    ):

        if (
            validation.get(
                field_name
            )
            is False
        ):

            cell48_dimension_failure_distribution[
                field_name
            ] += 1

    for reason in (
        validation.get(
            "failure_reasons",
            []
        )
        or []
    ):

        normalized_reason = (
            str(reason).strip()
        )

        if normalized_reason:

            cell48_failure_reason_distribution[
                normalized_reason
            ] += 1


# =========================================================
# STRUCTURAL FAILURE DIAGNOSTICS
# =========================================================

cell48_structural_error_distribution = Counter()


for result in cell48_full_semantic_results:

    if (
        result.get(
            "structural_status"
        )
        != "FAIL"
    ):
        continue

    errors = (
        result.get(
            "structural_errors"
        )
        or []
    )

    for error in errors:

        cell48_structural_error_distribution[
            str(error)
        ] += 1


# =========================================================
# COST
# =========================================================

cell48_known_costs = [
    result.get(
        "estimated_cost_toman"
    )
    for result
    in cell48_full_semantic_results
    if isinstance(
        result.get(
            "estimated_cost_toman"
        ),
        (int, float),
    )
]

cell48_total_estimated_cost_toman = (
    sum(
        cell48_known_costs
    )
    if cell48_known_costs
    else None
)


cell48_total_input_tokens = sum(
    int(
        (
            result.get(
                "usage"
            )
            or {}
        ).get(
            "input_tokens",
            0,
        )
        or 0
    )
    for result
    in cell48_full_semantic_results
)


cell48_total_output_tokens = sum(
    int(
        (
            result.get(
                "usage"
            )
            or {}
        ).get(
            "output_tokens",
            0,
        )
        or 0
    )
    for result
    in cell48_full_semantic_results
)


cell48_total_tokens = sum(
    int(
        (
            result.get(
                "usage"
            )
            or {}
        ).get(
            "total_tokens",
            0,
        )
        or 0
    )
    for result
    in cell48_full_semantic_results
)


# =========================================================
# IDENTITY
# =========================================================

cell48_result_candidate_ids = [
    result[
        "candidate_id"
    ]
    for result
    in cell48_full_semantic_results
]


cell48_identity_exact = (
    len(
        cell48_result_candidate_ids
    )
    == 3813
    and
    len(
        set(
            cell48_result_candidate_ids
        )
    )
    == 3813
    and
    set(
        cell48_result_candidate_ids
    )
    == set(
        cell48_candidate_ids
    )
)


# =========================================================
# RUNNER HEALTH
#
# Semantic FAIL is a data result,
# NOT a runner failure.
# =========================================================

cell48_runner_checks = {
    "3813_validation_requests_defined": (
        len(
            cell48_requests
        )
        == 3813
    ),

    "3813_results_reconstructed": (
        len(
            cell48_full_semantic_results
        )
        == 3813
    ),

    "3813_valid_checkpoints": (
        cell48_valid_checkpoint_count
        == 3813
    ),

    "zero_missing_checkpoints": (
        cell48_missing_checkpoint_count
        == 0
    ),

    "zero_invalid_checkpoints": (
        cell48_invalid_checkpoint_count
        == 0
    ),

    "exact_candidate_identity": (
        cell48_identity_exact
    ),

    "generation_population_preserved": (
        len(
            cell47_final_generation_results
        )
        == 3813
    ),

    "claim_evidence_revalidation_not_performed": True,
}


cell48_runner_status = (
    "PASS"
    if all(
        cell48_runner_checks.values()
    )
    else "FAIL"
)


cell48_structural_validation_status = (
    "PASS"
    if (
        cell48_structural_pass_count
        == 3813
        and
        cell48_api_failure_count
        == 0
    )
    else "FAIL"
)


if cell48_runner_status != "PASS":

    cell48_next_readiness = (
        "BLOCKED_CELL48_RUNNER_REPAIR_REQUIRED"
    )

elif cell48_api_failure_count > 0:

    cell48_next_readiness = (
        "READY_FOR_CELL48_API_FAILURE_RECOVERY"
    )

elif cell48_structural_fail_count > 0:

    cell48_next_readiness = (
        "READY_FOR_CELL48_STRUCTURAL_VALIDATOR_RECOVERY"
    )

elif cell48_semantic_fail_count > 0:

    cell48_next_readiness = (
        "READY_FOR_FULL_QA_SEMANTIC_FAILURE_ADJUDICATION"
    )

elif (
    cell48_semantic_pass_count
    == 3813
):

    cell48_next_readiness = (
        "READY_FOR_GOLDEN_V2_FINAL_RECORD_CONSTRUCTION"
    )

else:

    cell48_next_readiness = (
        "READY_FOR_CELL48_RESULT_ANALYSIS"
    )


# =========================================================
# OUTPUT
# =========================================================

print()

print(
    "=== GOLDEN V2 FULL QA SEMANTIC "
    "VALIDATION RESULT ==="
)

print()

print(
    "Target population:",
    len(
        cell48_requests
    ),
)

print(
    "Results reconstructed:",
    len(
        cell48_full_semantic_results
    ),
)

print(
    "Valid checkpoints:",
    cell48_valid_checkpoint_count,
)

print(
    "Missing checkpoints:",
    cell48_missing_checkpoint_count,
)

print(
    "Invalid checkpoints:",
    cell48_invalid_checkpoint_count,
)

print()

print(
    "API candidates this execution:",
    cell48_api_candidates_this_execution,
)

print(
    "API attempts this execution:",
    cell48_api_attempts_this_execution,
)

print(
    "Checkpoints created this execution:",
    cell48_checkpoints_created_this_execution,
)

print()

print(
    "Structural distribution:",
    dict(
        cell48_structural_distribution
    ),
)

print(
    "Structural PASS:",
    cell48_structural_pass_count,
    "/ 3813",
)

print(
    "Structural FAIL:",
    cell48_structural_fail_count,
    "/ 3813",
)

print(
    "API failures:",
    cell48_api_failure_count,
)

print(
    "Finish reason distribution:",
    dict(
        cell48_finish_reason_distribution
    ),
)

print(
    "Structural error distribution:",
    dict(
        cell48_structural_error_distribution
    ),
)

print()

print(
    "Semantic distribution:",
    dict(
        cell48_semantic_distribution
    ),
)

print(
    "Semantic PASS:",
    cell48_semantic_pass_count,
)

print(
    "Semantic FAIL:",
    cell48_semantic_fail_count,
)

print()

print(
    "Dimension failure distribution:",
    dict(
        cell48_dimension_failure_distribution
    ),
)

print(
    "Failure reason distribution:",
    dict(
        cell48_failure_reason_distribution
    ),
)

print()

print(
    "Input tokens represented by checkpoints:",
    cell48_total_input_tokens,
)

print(
    "Output tokens represented by checkpoints:",
    cell48_total_output_tokens,
)

print(
    "Total tokens represented by checkpoints:",
    cell48_total_tokens,
)

print(
    "Estimated semantic-validation cost "
    "represented by checkpoints (toman):",
    cell48_total_estimated_cost_toman,
)

print()

print(
    "Configured gateway limit/minute:",
    120,
)

print(
    "Configured request starts/minute:",
    round(
        60.0
        / CELL48_REQUEST_START_SPACING_SECONDS,
        2,
    ),
)

print(
    "Configured request timeout seconds:",
    CELL48_REQUEST_TIMEOUT_SECONDS,
)

print()

print("Runner checks:")

for (
    check_name,
    check_value,
) in cell48_runner_checks.items():

    print(
        " -",
        check_name,
        ":",
        (
            "PASS"
            if check_value
            else "FAIL"
        ),
    )

print()

print(
    "Runner health:",
    cell48_runner_status,
)

print(
    "Structural validation result:",
    cell48_structural_validation_status,
)

print(
    "Next readiness:",
    cell48_next_readiness,
)

print(
    "CELL 48 CHECKPOINTED RUNNER:",
    cell48_runner_status,
)


assert cell48_runner_status == "PASS", (
    "Cell 48 checkpointed runner health failed."
)
````

<div style="direction: rtl; white-space: normal; line-height: 1.7;">
<h2>مرحله ۴۹ — بازیابی هدفمند Failureهای فنی اعتبارسنجی معنایی کامل</h2>
<p>
در مرحله قبل، اعتبارسنجی معنایی کامل برای هر ۳۸۱۳ Question/Answer دارای Checkpoint معتبر شد و Runner با موفقیت پایان یافت؛ با این حال ۹۸ مورد به دلیل Failure فنی هنوز Judgment معنایی قابل استفاده ندارند. از این تعداد، ۹۰ مورد با وضعیت API_FAILED و ۸ مورد به دلیل پایان یافتن پاسخ با finish_reason=length و شکست در Parse ساختاری باقی مانده‌اند.
</p>
<p>
در این مرحله فقط همین ۹۸ Candidate بازیابی می‌شوند و ۳۷۱۵ نتیجه ساختاری سالم مرحله ۴۸ بدون هیچ فراخوانی مجدد به مدل حفظ خواهند شد. Recovery در Namespace مستقل و با Fingerprint مستقل انجام می‌شود تا Checkpointهای مرحله ۴۸ تغییر نکنند و اجرای مجدد نیز فقط موارد فاقد Recovery Checkpoint معتبر را ارسال کند. برای کاهش Failure ناشی از طول پاسخ، سقف خروجی Validator در Recovery افزایش می‌یابد، در حالی که Protocol معنایی، Prompt اعتبارسنجی و ابعاد هشت‌گانه ارزیابی بدون تغییر باقی می‌مانند.
</p>
<p>
معیار موفقیت این مرحله این است که تمام ۹۸ Failure فنی تعیین‌تکلیف شوند، هیچ Candidate سالم مرحله ۴۸ دوباره به API ارسال نشود، هویت کامل جمعیت ۳۸۱۳ مورد حفظ شود و یک مجموعه نهایی از Judgmentهای ساختاری معتبر برای ورود به مرحله رسیدگی به Semantic FAILها ساخته شود.
</p>
</div>

````python
# Code Cell 49 — Targeted Recovery of Cell 48 Technical Validation Failures

from collections import Counter
from concurrent.futures import ThreadPoolExecutor, as_completed
from pathlib import Path
import hashlib
import json
import os
import re
import tempfile
import threading
import time


# =========================================================
# CONFIG
# =========================================================

CELL49_RUNNER_VERSION = (
    "golden_v2_qa_semantic_technical_recovery_v1"
)

CELL49_VALIDATION_PROTOCOL_VERSION = (
    CELL48_VALIDATION_PROTOCOL_VERSION
)

CELL49_OUTPUT_SCHEMA_VERSION = (
    CELL48_OUTPUT_SCHEMA_VERSION
)

# Same semantic validator; only output capacity is increased.
CELL49_MAX_TOKENS = 3072
CELL49_TEMPERATURE = 0.0
CELL49_TOP_P = 1.0

CELL49_MAX_ATTEMPTS = 2
CELL49_RETRY_SECONDS = 2.0

CELL49_MAX_WORKERS = 16
CELL49_REQUEST_START_SPACING_SECONDS = 0.6
CELL49_REQUEST_TIMEOUT_SECONDS = 180.0

# Paid execution is allowed only for missing exact Cell 49
# recovery checkpoints. Before final Restart -> Run All,
# this flag must be changed to False.
CELL49_ALLOW_API_FOR_MISSING = False


# =========================================================
# REQUIRED STATE
# =========================================================

cell49_required_names = [
    "project_root",
    "semantic_review_client",
    "semantic_review_model",
    "cell25_candidate_skeletons",
    "cell47_final_generation_results",
    "cell48_full_semantic_results",
    "CELL37_SYSTEM_PROMPT",
    "CELL37_BOOLEAN_FIELDS",
    "cell37_validate_output",
]

cell49_missing_state = [
    name
    for name in cell49_required_names
    if name not in globals()
]

assert not cell49_missing_state, (
    "Missing required Cell 49 state: "
    + ", ".join(cell49_missing_state)
)


# =========================================================
# CELL 48 SOURCE VALIDATION
# =========================================================

assert len(cell48_full_semantic_results) == 3813, (
    "Cell 49 requires exactly 3813 Cell 48 results."
)

assert len({
    result["candidate_id"]
    for result in cell48_full_semantic_results
}) == 3813, (
    "Cell 48 candidate identity is not unique."
)

cell49_cell48_by_candidate_id = {
    result["candidate_id"]: result
    for result in cell48_full_semantic_results
}

cell49_skeleton_by_candidate_id = {
    record["candidate_id"]: record
    for record in cell25_candidate_skeletons
}

cell49_generation_by_candidate_id = {
    result["candidate_id"]: result
    for result in cell47_final_generation_results
}

assert (
    set(cell49_cell48_by_candidate_id)
    == set(cell49_skeleton_by_candidate_id)
    == set(cell49_generation_by_candidate_id)
), (
    "Candidate identity mismatch across Cell 25, "
    "Cell 47 and Cell 48."
)


# =========================================================
# EXACT TECHNICAL-FAILURE POPULATION
# =========================================================

cell49_preserved_results = []
cell49_recovery_source_results = []

for result in cell48_full_semantic_results:

    if result.get("structural_status") == "PASS":

        cell49_preserved_results.append(
            result
        )

    else:

        cell49_recovery_source_results.append(
            result
        )


assert len(cell49_preserved_results) == 3715, (
    "Expected exactly 3715 structurally valid "
    "Cell 48 results to preserve."
)

assert len(cell49_recovery_source_results) == 98, (
    "Expected exactly 98 Cell 48 technical failures "
    "for targeted recovery."
)


cell49_api_failed_source_results = [
    result
    for result in cell49_recovery_source_results
    if result.get("execution_status") == "API_FAILED"
]

cell49_length_source_results = [
    result
    for result in cell49_recovery_source_results
    if (
        result.get("finish_reason") == "length"
        and
        result.get("execution_status") != "API_FAILED"
    )
]


assert len(cell49_api_failed_source_results) == 90, (
    "Expected exactly 90 API_FAILED Cell 48 results."
)

assert len(cell49_length_source_results) == 8, (
    "Expected exactly 8 length-truncated Cell 48 results."
)


cell49_recovery_candidate_ids = [
    result["candidate_id"]
    for result in cell49_recovery_source_results
]

assert len(cell49_recovery_candidate_ids) == 98
assert len(set(cell49_recovery_candidate_ids)) == 98


# =========================================================
# HASH UTILITIES
# =========================================================

def cell49_canonical_json(value):

    return json.dumps(
        value,
        ensure_ascii=False,
        sort_keys=True,
        separators=(",", ":"),
    )


def cell49_sha256_text(text):

    return hashlib.sha256(
        text.encode("utf-8")
    ).hexdigest()


# =========================================================
# VALIDATION PACKET
# =========================================================

def cell49_build_validation_packet(candidate_id):

    skeleton = (
        cell49_skeleton_by_candidate_id[
            candidate_id
        ]
    )

    generation_result = (
        cell49_generation_by_candidate_id[
            candidate_id
        ]
    )

    generated_qa = generation_result.get(
        "parsed_json"
    )

    assert isinstance(
        generated_qa,
        dict,
    ), (
        f"Generated QA missing for {candidate_id}"
    )

    evidence_ids = skeleton[
        "evidence_ids"
    ]

    evidence_quotes = skeleton[
        "evidence_quotes"
    ]

    assert len(evidence_ids) == len(evidence_quotes), (
        f"Evidence identity mismatch for {candidate_id}"
    )

    evidence = []

    for index, (
        evidence_id,
        evidence_quote,
    ) in enumerate(
        zip(
            evidence_ids,
            evidence_quotes,
        ),
        start=1,
    ):

        evidence.append({
            "index": index,
            "evidence_id": evidence_id,
            "quote": evidence_quote,
        })

    return {
        "candidate_id": candidate_id,

        "claim": {
            "claim_id": skeleton[
                "claim_id"
            ],

            "claim_text_fa": skeleton[
                "claim_text_fa"
            ],

            "claim_type": skeleton.get(
                "claim_type"
            ),

            "claim_types": skeleton.get(
                "claim_types",
                [],
            ),

            "authority_class": skeleton.get(
                "authority_class"
            ),
        },

        "evidence": evidence,

        "generated_qa": {
            "question": generated_qa[
                "question"
            ],

            "expected_answer": generated_qa[
                "expected_answer"
            ],

            "question_intent": generated_qa[
                "question_intent"
            ],
        },
    }


# =========================================================
# EXACT SAME SEMANTIC TASK AS CELL 48
# =========================================================

def cell49_build_user_prompt(candidate_id):

    packet = (
        cell49_build_validation_packet(
            candidate_id
        )
    )

    return (
        "Validate ONLY the generated Question and Expected Answer "
        "against the supplied prevalidated Claim and Evidence.\n\n"

        "The Claim and Evidence have already passed upstream semantic "
        "validation. Do NOT decide whether the Claim itself should be "
        "accepted or rejected. Do NOT perform a new Claim-Evidence "
        "eligibility review.\n\n"

        "Evaluate exactly these QA dimensions:\n"
        "1. question_targets_claim\n"
        "2. answer_recovers_claim\n"
        "3. scope_fidelity\n"
        "4. modality_quantifier_fidelity\n"
        "5. standalone_question\n"
        "6. no_neighbor_fact_drift\n"
        "7. no_external_knowledge\n"
        "8. answer_question_alignment\n\n"

        "Important fidelity rules:\n"
        "- The Question must ask for information actually asserted "
        "by the Claim.\n"
        "- The Expected Answer must correctly answer that Question "
        "and recover the relevant meaning of the Claim.\n"
        "- Preserve meaning-bearing negation, modality, attribution, "
        "conditions, exceptions, scope, quantities, percentages, "
        "comparisons and quantifiers such as only/تنها/فقط when "
        "they are relevant to the generated QA target.\n"
        "- Do not permit WHY/HOW/cause/reason/mechanism claims unless "
        "the supplied Claim actually asserts them.\n"
        "- Do not allow facts imported from neighboring Evidence text "
        "when those facts are not part of the target Claim.\n"
        "- Do not require external knowledge.\n"
        "- A narrower Question/Answer may be valid only when it "
        "faithfully targets a self-contained asserted part of the "
        "Claim without changing its meaning or introducing drift.\n\n"

        "Return only the JSON object required by the validation "
        "system prompt.\n\n"

        "VALIDATION PACKET:\n"
        + json.dumps(
            packet,
            ensure_ascii=False,
            sort_keys=True,
            indent=2,
        )
    )


# =========================================================
# RECOVERY REQUEST CONTRACT
# =========================================================

def cell49_build_request(candidate_id):

    packet = (
        cell49_build_validation_packet(
            candidate_id
        )
    )

    user_prompt = (
        cell49_build_user_prompt(
            candidate_id
        )
    )

    generated_qa = (
        cell49_generation_by_candidate_id[
            candidate_id
        ]["parsed_json"]
    )

    generated_qa_sha256 = (
        cell49_sha256_text(
            cell49_canonical_json(
                generated_qa
            )
        )
    )

    validation_packet_sha256 = (
        cell49_sha256_text(
            cell49_canonical_json(
                packet
            )
        )
    )

    source_cell48_result = (
        cell49_cell48_by_candidate_id[
            candidate_id
        ]
    )

    source_failure_contract = {
        "execution_status": (
            source_cell48_result.get(
                "execution_status"
            )
        ),

        "structural_status": (
            source_cell48_result.get(
                "structural_status"
            )
        ),

        "finish_reason": (
            source_cell48_result.get(
                "finish_reason"
            )
        ),

        "structural_errors": (
            source_cell48_result.get(
                "structural_errors",
                [],
            )
        ),
    }

    contract = {
        "runner_version": (
            CELL49_RUNNER_VERSION
        ),

        "validation_protocol_version": (
            CELL49_VALIDATION_PROTOCOL_VERSION
        ),

        "output_schema_version": (
            CELL49_OUTPUT_SCHEMA_VERSION
        ),

        "candidate_id": candidate_id,

        "generated_qa_sha256": (
            generated_qa_sha256
        ),

        "validation_packet_sha256": (
            validation_packet_sha256
        ),

        "source_cell48_failure": (
            source_failure_contract
        ),

        "model": semantic_review_model,

        "system_prompt": (
            CELL37_SYSTEM_PROMPT
        ),

        "user_prompt": user_prompt,

        "temperature": (
            CELL49_TEMPERATURE
        ),

        "top_p": (
            CELL49_TOP_P
        ),

        "max_tokens": (
            CELL49_MAX_TOKENS
        ),

        "enable_thinking": False,
    }

    request_fingerprint = (
        cell49_sha256_text(
            cell49_canonical_json(
                contract
            )
        )
    )

    return {
        **contract,

        "request_fingerprint": (
            request_fingerprint
        ),
    }


cell49_requests = [
    cell49_build_request(
        candidate_id
    )
    for candidate_id
    in cell49_recovery_candidate_ids
]

assert len(cell49_requests) == 98

assert len({
    request["candidate_id"]
    for request in cell49_requests
}) == 98

assert len({
    request["request_fingerprint"]
    for request in cell49_requests
}) == 98


# =========================================================
# INDEPENDENT RECOVERY CHECKPOINT NAMESPACE
# =========================================================

cell49_checkpoint_dir = (
    Path(project_root)
    / "artifacts"
    / "phase2"
    / "golden_dataset_v2"
    / "work"
    / "qa_semantic_technical_recovery_v1"
)

cell49_checkpoint_dir.mkdir(
    parents=True,
    exist_ok=True,
)


def cell49_checkpoint_path(candidate_id):

    safe_id = re.sub(
        r"[^A-Za-z0-9_.-]+",
        "_",
        candidate_id,
    )

    return (
        cell49_checkpoint_dir
        / f"{safe_id}.json"
    )


# =========================================================
# ATOMIC CHECKPOINT WRITE
# =========================================================

def cell49_atomic_write_json(
    path,
    payload,
):

    path = Path(path)

    content = (
        json.dumps(
            payload,
            ensure_ascii=False,
            sort_keys=True,
            indent=2,
        )
        + "\n"
    )

    fd, temp_name = tempfile.mkstemp(
        prefix=path.name + ".",
        suffix=".tmp",
        dir=str(path.parent),
    )

    try:

        with os.fdopen(
            fd,
            "w",
            encoding="utf-8",
        ) as handle:

            handle.write(content)
            handle.flush()
            os.fsync(
                handle.fileno()
            )

        os.replace(
            temp_name,
            path,
        )

    except Exception:

        try:

            os.unlink(
                temp_name
            )

        except FileNotFoundError:

            pass

        raise


# =========================================================
# CHECKPOINT CONTRACT
# =========================================================

def cell49_build_checkpoint(
    request,
    result,
):

    checkpoint = {
        "checkpoint_version": (
            CELL49_RUNNER_VERSION
        ),

        "candidate_id": (
            request["candidate_id"]
        ),

        "request_fingerprint": (
            request[
                "request_fingerprint"
            ]
        ),

        "generated_qa_sha256": (
            request[
                "generated_qa_sha256"
            ]
        ),

        "validation_packet_sha256": (
            request[
                "validation_packet_sha256"
            ]
        ),

        "validation_protocol_version": (
            CELL49_VALIDATION_PROTOCOL_VERSION
        ),

        "output_schema_version": (
            CELL49_OUTPUT_SCHEMA_VERSION
        ),

        "model": (
            semantic_review_model
        ),

        "temperature": (
            CELL49_TEMPERATURE
        ),

        "top_p": (
            CELL49_TOP_P
        ),

        "max_tokens": (
            CELL49_MAX_TOKENS
        ),

        "result": result,
    }

    checkpoint[
        "checkpoint_payload_sha256"
    ] = cell49_sha256_text(
        cell49_canonical_json(
            checkpoint
        )
    )

    return checkpoint


def cell49_validate_checkpoint(
    checkpoint,
    request,
):

    errors = []

    if not isinstance(
        checkpoint,
        dict,
    ):

        return [
            "CHECKPOINT_NOT_OBJECT"
        ]

    stored_sha = checkpoint.get(
        "checkpoint_payload_sha256"
    )

    payload_without_sha = {
        key: value
        for key, value
        in checkpoint.items()
        if key
        != "checkpoint_payload_sha256"
    }

    recalculated_sha = (
        cell49_sha256_text(
            cell49_canonical_json(
                payload_without_sha
            )
        )
    )

    if stored_sha != recalculated_sha:

        errors.append(
            "CHECKPOINT_HASH_MISMATCH"
        )

    expected_values = {
        "checkpoint_version": (
            CELL49_RUNNER_VERSION
        ),

        "candidate_id": (
            request["candidate_id"]
        ),

        "request_fingerprint": (
            request[
                "request_fingerprint"
            ]
        ),

        "generated_qa_sha256": (
            request[
                "generated_qa_sha256"
            ]
        ),

        "validation_packet_sha256": (
            request[
                "validation_packet_sha256"
            ]
        ),

        "validation_protocol_version": (
            CELL49_VALIDATION_PROTOCOL_VERSION
        ),

        "output_schema_version": (
            CELL49_OUTPUT_SCHEMA_VERSION
        ),

        "model": (
            semantic_review_model
        ),

        "temperature": (
            CELL49_TEMPERATURE
        ),

        "top_p": (
            CELL49_TOP_P
        ),

        "max_tokens": (
            CELL49_MAX_TOKENS
        ),
    }

    for (
        field_name,
        expected_value,
    ) in expected_values.items():

        if (
            checkpoint.get(
                field_name
            )
            != expected_value
        ):

            errors.append(
                f"{field_name.upper()}_MISMATCH"
            )

    result = checkpoint.get(
        "result"
    )

    if not isinstance(
        result,
        dict,
    ):

        errors.append(
            "CHECKPOINT_RESULT_INVALID"
        )

    elif (
        result.get(
            "candidate_id"
        )
        != request["candidate_id"]
    ):

        errors.append(
            "CHECKPOINT_RESULT_CANDIDATE_MISMATCH"
        )

    return sorted(
        set(errors)
    )


def cell49_load_checkpoint(request):

    path = cell49_checkpoint_path(
        request["candidate_id"]
    )

    if not path.exists():

        return (
            None,
            "MISSING",
            [],
        )

    try:

        with path.open(
            "r",
            encoding="utf-8",
        ) as handle:

            checkpoint = json.load(
                handle
            )

    except Exception as exc:

        return (
            None,
            "UNREADABLE",
            [
                f"{type(exc).__name__}: {exc}"
            ],
        )

    errors = (
        cell49_validate_checkpoint(
            checkpoint,
            request,
        )
    )

    if errors:

        return (
            checkpoint,
            "INVALID",
            errors,
        )

    return (
        checkpoint,
        "VALID",
        [],
    )


# =========================================================
# RESPONSE UTILITIES
# =========================================================

def cell49_clean_model_json(content):

    if not isinstance(
        content,
        str,
    ):

        return ""

    clean = content.strip()

    if clean.startswith("```"):

        clean = re.sub(
            r"^```(?:json)?\s*",
            "",
            clean,
            flags=re.IGNORECASE,
        )

        clean = re.sub(
            r"\s*```$",
            "",
            clean,
        )

    return clean.strip()


def cell49_is_rate_limit_exception(exc):

    error_text = str(
        exc
    ).lower()

    error_type = type(
        exc
    ).__name__.lower()

    return (
        "ratelimit" in error_type
        or
        "rate_limit" in error_type
        or
        "429" in error_text
        or
        "too many requests" in error_text
    )


# =========================================================
# GLOBAL REQUEST START THROTTLE
# =========================================================

cell49_request_start_lock = (
    threading.Lock()
)

cell49_last_request_start = [
    0.0
]


def cell49_wait_for_request_slot():

    with cell49_request_start_lock:

        now = time.monotonic()

        elapsed = (
            now
            - cell49_last_request_start[0]
        )

        wait_seconds = max(
            0.0,
            (
                CELL49_REQUEST_START_SPACING_SECONDS
                - elapsed
            ),
        )

        if wait_seconds > 0:

            time.sleep(
                wait_seconds
            )

        cell49_last_request_start[0] = (
            time.monotonic()
        )


# =========================================================
# COST
# =========================================================

cell49_has_cost_rates = (
    "INPUT_TOMAN_PER_TOKEN"
    in globals()
    and
    "OUTPUT_TOMAN_PER_TOKEN"
    in globals()
)


def cell49_estimated_cost_toman(
    input_tokens,
    output_tokens,
):

    if not cell49_has_cost_rates:

        return None

    return (
        input_tokens
        * INPUT_TOMAN_PER_TOKEN
        +
        output_tokens
        * OUTPUT_TOMAN_PER_TOKEN
    )


# =========================================================
# FAILURE RESULT
# =========================================================

def cell49_build_api_failure_result(
    request,
    api_attempts,
    attempt_history,
    elapsed_seconds=None,
):

    return {
        "candidate_id": (
            request["candidate_id"]
        ),

        "request_fingerprint": (
            request[
                "request_fingerprint"
            ]
        ),

        "validation_protocol_version": (
            CELL49_VALIDATION_PROTOCOL_VERSION
        ),

        "execution_status": (
            "API_FAILED"
        ),

        "api_attempts": (
            api_attempts
        ),

        "attempt_history": (
            attempt_history
        ),

        "finish_reason": None,

        "elapsed_seconds": (
            round(
                elapsed_seconds,
                4,
            )
            if isinstance(
                elapsed_seconds,
                (int, float),
            )
            else None
        ),

        "usage": {
            "input_tokens": 0,
            "output_tokens": 0,
            "reasoning_tokens": 0,
            "total_tokens": 0,
        },

        "estimated_cost_toman": None,

        "raw_content": None,

        "parsed_json": None,

        "parse_error": None,

        "structural_errors": [
            "API_FAILED"
        ],

        "structural_status": (
            "FAIL"
        ),

        "semantic_verdict": None,

        "validation": None,
    }


# =========================================================
# TARGETED VALIDATOR EXECUTION
# =========================================================

def cell49_execute_request(request):

    attempt_history = []
    api_attempts = 0

    for attempt_number in range(
        1,
        CELL49_MAX_ATTEMPTS + 1,
    ):

        cell49_wait_for_request_slot()

        started_at = time.time()
        api_attempts += 1

        try:

            response = (
                semantic_review_client
                .chat
                .completions
                .create(
                    model=(
                        semantic_review_model
                    ),

                    messages=[
                        {
                            "role": "system",
                            "content": (
                                CELL37_SYSTEM_PROMPT
                            ),
                        },
                        {
                            "role": "user",
                            "content": (
                                request[
                                    "user_prompt"
                                ]
                            ),
                        },
                    ],

                    temperature=(
                        CELL49_TEMPERATURE
                    ),

                    top_p=(
                        CELL49_TOP_P
                    ),

                    max_tokens=(
                        CELL49_MAX_TOKENS
                    ),

                    timeout=(
                        CELL49_REQUEST_TIMEOUT_SECONDS
                    ),

                    extra_body={
                        "enable_thinking": False,
                    },
                )
            )

            if response is None:

                raise RuntimeError(
                    "EMPTY_API_RESPONSE"
                )

            response_choices = getattr(
                response,
                "choices",
                None,
            )

            if (
                not isinstance(
                    response_choices,
                    (list, tuple),
                )
                or
                len(response_choices) == 0
            ):

                raise RuntimeError(
                    "API_RESPONSE_CHOICES_MISSING_OR_EMPTY"
                )

            choice = (
                response_choices[0]
            )

            if choice is None:

                raise RuntimeError(
                    "API_RESPONSE_FIRST_CHOICE_IS_NONE"
                )

            message = getattr(
                choice,
                "message",
                None,
            )

            if message is None:

                raise RuntimeError(
                    "API_RESPONSE_MESSAGE_MISSING"
                )

        except Exception as exc:

            elapsed_seconds = (
                time.time()
                - started_at
            )

            attempt_history.append({
                "attempt": (
                    attempt_number
                ),

                "status": (
                    "API_ERROR"
                ),

                "error": (
                    f"{type(exc).__name__}: {exc}"
                ),
            })

            if (
                attempt_number
                < CELL49_MAX_ATTEMPTS
            ):

                wait_seconds = (
                    70.0
                    if cell49_is_rate_limit_exception(
                        exc
                    )
                    else (
                        CELL49_RETRY_SECONDS
                        * attempt_number
                    )
                )

                time.sleep(
                    wait_seconds
                )

                continue

            return (
                cell49_build_api_failure_result(
                    request=request,
                    api_attempts=api_attempts,
                    attempt_history=attempt_history,
                    elapsed_seconds=elapsed_seconds,
                )
            )

        elapsed_seconds = (
            time.time()
            - started_at
        )

        finish_reason = getattr(
            choice,
            "finish_reason",
            None,
        )

        raw_content = getattr(
            message,
            "content",
            None,
        )

        usage = getattr(
            response,
            "usage",
            None,
        )

        input_tokens = int(
            getattr(
                usage,
                "prompt_tokens",
                0,
            )
            or 0
        )

        output_tokens = int(
            getattr(
                usage,
                "completion_tokens",
                0,
            )
            or 0
        )

        total_tokens = int(
            getattr(
                usage,
                "total_tokens",
                0,
            )
            or (
                input_tokens
                + output_tokens
            )
        )

        reasoning_tokens = 0

        completion_details = getattr(
            usage,
            "completion_tokens_details",
            None,
        )

        if completion_details is not None:

            reasoning_tokens = int(
                getattr(
                    completion_details,
                    "reasoning_tokens",
                    0,
                )
                or 0
            )

        clean_content = (
            cell49_clean_model_json(
                raw_content
            )
        )

        parsed_json = None
        parse_error = None

        try:

            parsed_json = json.loads(
                clean_content
            )

        except Exception as exc:

            parse_error = (
                f"{type(exc).__name__}: {exc}"
            )

        if parsed_json is None:

            structural_errors = [
                "JSON_PARSE_FAILED"
            ]

        else:

            structural_errors = (
                cell37_validate_output(
                    parsed_json
                )
            )

        structural_status = (
            "PASS"
            if not structural_errors
            else "FAIL"
        )

        semantic_verdict = None

        if (
            structural_status
            == "PASS"
        ):

            semantic_verdict = (
                parsed_json.get(
                    "verdict"
                )
            )

        return {
            "candidate_id": (
                request["candidate_id"]
            ),

            "request_fingerprint": (
                request[
                    "request_fingerprint"
                ]
            ),

            "validation_protocol_version": (
                CELL49_VALIDATION_PROTOCOL_VERSION
            ),

            "execution_status": (
                "COMPLETED"
            ),

            "api_attempts": (
                api_attempts
            ),

            "attempt_history": (
                attempt_history
            ),

            "finish_reason": (
                finish_reason
            ),

            "elapsed_seconds": round(
                elapsed_seconds,
                4,
            ),

            "usage": {
                "input_tokens": (
                    input_tokens
                ),

                "output_tokens": (
                    output_tokens
                ),

                "reasoning_tokens": (
                    reasoning_tokens
                ),

                "total_tokens": (
                    total_tokens
                ),
            },

            "estimated_cost_toman": (
                cell49_estimated_cost_toman(
                    input_tokens,
                    output_tokens,
                )
            ),

            "raw_content": (
                raw_content
            ),

            "parsed_json": (
                parsed_json
            ),

            "parse_error": (
                parse_error
            ),

            "structural_errors": (
                structural_errors
            ),

            "structural_status": (
                structural_status
            ),

            "semantic_verdict": (
                semantic_verdict
            ),

            "validation": (
                parsed_json
                if structural_status
                == "PASS"
                else None
            ),
        }

    raise AssertionError(
        "Unexpected Cell 49 execution path."
    )


# =========================================================
# PREFLIGHT CHECKPOINT SCAN
# =========================================================

cell49_valid_before_api = {}
cell49_missing_requests = []
cell49_invalid_before_api = []


for request in cell49_requests:

    (
        checkpoint,
        status,
        errors,
    ) = cell49_load_checkpoint(
        request
    )

    candidate_id = (
        request["candidate_id"]
    )

    if status == "VALID":

        cell49_valid_before_api[
            candidate_id
        ] = checkpoint

    elif status == "MISSING":

        cell49_missing_requests.append(
            request
        )

    else:

        cell49_invalid_before_api.append({
            "candidate_id": (
                candidate_id
            ),

            "status": status,

            "errors": errors,
        })


assert not cell49_invalid_before_api, (
    "Invalid/unreadable Cell 49 checkpoints detected. "
    "Refusing automatic paid replacement. "
    f"Examples: {cell49_invalid_before_api[:5]}"
)


# =========================================================
# PREFLIGHT OUTPUT
# =========================================================

print(
    "=== GOLDEN V2 TARGETED SEMANTIC "
    "TECHNICAL RECOVERY PREFLIGHT ==="
)

print()

print(
    "Full Cell 48 population:",
    len(
        cell48_full_semantic_results
    ),
)

print(
    "Preserved structural PASS:",
    len(
        cell49_preserved_results
    ),
)

print(
    "Recovery population:",
    len(
        cell49_requests
    ),
)

print(
    "Source API_FAILED:",
    len(
        cell49_api_failed_source_results
    ),
)

print(
    "Source length failures:",
    len(
        cell49_length_source_results
    ),
)

print(
    "Exact valid recovery checkpoints:",
    len(
        cell49_valid_before_api
    ),
)

print(
    "Missing recovery checkpoints:",
    len(
        cell49_missing_requests
    ),
)

print(
    "Model:",
    semantic_review_model,
)

print(
    "Validation protocol:",
    CELL49_VALIDATION_PROTOCOL_VERSION,
)

print(
    "Recovery max tokens:",
    CELL49_MAX_TOKENS,
)

print(
    "Parallel workers:",
    CELL49_MAX_WORKERS,
)

print(
    "Request start spacing seconds:",
    CELL49_REQUEST_START_SPACING_SECONDS,
)

print(
    "Request timeout seconds:",
    CELL49_REQUEST_TIMEOUT_SECONDS,
)

print(
    "API allowed for missing exact recovery checkpoints:",
    CELL49_ALLOW_API_FOR_MISSING,
)

print(
    "Previously valid Cell 48 results sent to API:",
    0,
)

print()


if (
    cell49_missing_requests
    and
    not CELL49_ALLOW_API_FOR_MISSING
):

    raise AssertionError(
        "Cell 49 has missing recovery checkpoints "
        "but API execution is disabled."
    )


# =========================================================
# TARGETED RECOVERY EXECUTION
# =========================================================

cell49_api_candidates_this_execution = 0
cell49_api_attempts_this_execution = 0
cell49_checkpoints_created_this_execution = 0


def cell49_execute_and_checkpoint(
    request,
):

    candidate_id = (
        request["candidate_id"]
    )

    result = (
        cell49_execute_request(
            request
        )
    )

    checkpoint = (
        cell49_build_checkpoint(
            request,
            result,
        )
    )

    cell49_atomic_write_json(
        cell49_checkpoint_path(
            candidate_id
        ),
        checkpoint,
    )

    (
        verified_checkpoint,
        verified_status,
        verified_errors,
    ) = cell49_load_checkpoint(
        request
    )

    if verified_status != "VALID":

        raise AssertionError(
            "Cell 49 checkpoint verification "
            f"failed for {candidate_id}: "
            f"{verified_errors}"
        )

    return (
        candidate_id,
        verified_checkpoint[
            "result"
        ],
    )


if cell49_missing_requests:

    completed_count = 0

    executor = ThreadPoolExecutor(
        max_workers=(
            CELL49_MAX_WORKERS
        )
    )

    future_to_request = {}

    try:

        future_to_request = {
            executor.submit(
                cell49_execute_and_checkpoint,
                request,
            ): request

            for request
            in cell49_missing_requests
        }

        for future in as_completed(
            future_to_request
        ):

            request = (
                future_to_request[
                    future
                ]
            )

            candidate_id = (
                request[
                    "candidate_id"
                ]
            )

            try:

                (
                    resolved_candidate_id,
                    result,
                ) = future.result()

            except Exception as exc:

                completed_count += 1

                print(
                    f"[{completed_count:03d}/"
                    f"{len(cell49_missing_requests):03d}]",
                    candidate_id,
                    "| WORKER_EXCEPTION =",
                    f"{type(exc).__name__}: {exc}",
                    flush=True,
                )

                continue

            assert (
                resolved_candidate_id
                == candidate_id
            )

            completed_count += 1

            cell49_api_candidates_this_execution += 1

            cell49_api_attempts_this_execution += int(
                result.get(
                    "api_attempts",
                    0,
                )
                or 0
            )

            cell49_checkpoints_created_this_execution += 1

            print(
                f"[{completed_count:03d}/"
                f"{len(cell49_missing_requests):03d}]",
                candidate_id,
                "| structural=",
                result.get(
                    "structural_status"
                ),
                "| semantic=",
                result.get(
                    "semantic_verdict"
                ),
                "| finish=",
                result.get(
                    "finish_reason"
                ),
                flush=True,
            )

    except KeyboardInterrupt:

        print(
            "\nCell 49 interrupted by user. "
            "Completed recovery checkpoints remain preserved.",
            flush=True,
        )

        for future in future_to_request:

            if not future.done():

                future.cancel()

        raise

    finally:

        executor.shutdown(
            wait=False,
            cancel_futures=True,
        )


# =========================================================
# FINAL RECOVERY CHECKPOINT RESCAN
# =========================================================

cell49_recovery_results = []

cell49_valid_checkpoint_count = 0
cell49_missing_checkpoint_count = 0
cell49_invalid_checkpoint_count = 0


for request in cell49_requests:

    (
        checkpoint,
        status,
        errors,
    ) = cell49_load_checkpoint(
        request
    )

    if status == "VALID":

        cell49_valid_checkpoint_count += 1

        cell49_recovery_results.append(
            checkpoint[
                "result"
            ]
        )

    elif status == "MISSING":

        cell49_missing_checkpoint_count += 1

    else:

        cell49_invalid_checkpoint_count += 1


# =========================================================
# RECOVERY RESULT DISTRIBUTIONS
# =========================================================

cell49_recovery_structural_distribution = Counter(
    result.get(
        "structural_status"
    )
    for result
    in cell49_recovery_results
)

cell49_recovery_execution_distribution = Counter(
    result.get(
        "execution_status"
    )
    for result
    in cell49_recovery_results
)

cell49_recovery_finish_distribution = Counter(
    result.get(
        "finish_reason"
    )
    for result
    in cell49_recovery_results
)

cell49_recovery_semantic_distribution = Counter(
    result.get(
        "semantic_verdict"
    )
    for result
    in cell49_recovery_results
    if result.get(
        "structural_status"
    )
    == "PASS"
)


# =========================================================
# MERGE PRESERVED CELL 48 + CELL 49 RECOVERY
# =========================================================

cell49_recovery_by_candidate_id = {
    result["candidate_id"]: result
    for result in cell49_recovery_results
}

cell49_final_semantic_results = []

for candidate_id in [
    record["candidate_id"]
    for record in cell25_candidate_skeletons
]:

    if candidate_id in (
        cell49_recovery_by_candidate_id
    ):

        cell49_final_semantic_results.append(
            cell49_recovery_by_candidate_id[
                candidate_id
            ]
        )

    else:

        cell49_final_semantic_results.append(
            cell49_cell48_by_candidate_id[
                candidate_id
            ]
        )


# =========================================================
# FINAL 3813-WIDE STATE
# =========================================================

cell49_final_structural_distribution = Counter(
    result.get(
        "structural_status"
    )
    for result
    in cell49_final_semantic_results
)

cell49_final_execution_distribution = Counter(
    result.get(
        "execution_status"
    )
    for result
    in cell49_final_semantic_results
)

cell49_final_semantic_distribution = Counter(
    result.get(
        "semantic_verdict"
    )
    for result
    in cell49_final_semantic_results
    if result.get(
        "structural_status"
    )
    == "PASS"
)


cell49_final_structural_pass_count = (
    cell49_final_structural_distribution.get(
        "PASS",
        0,
    )
)

cell49_final_structural_fail_count = (
    cell49_final_structural_distribution.get(
        "FAIL",
        0,
    )
)

cell49_final_api_failure_count = (
    cell49_final_execution_distribution.get(
        "API_FAILED",
        0,
    )
)

cell49_final_semantic_pass_count = (
    cell49_final_semantic_distribution.get(
        "PASS",
        0,
    )
)

cell49_final_semantic_fail_count = (
    cell49_final_semantic_distribution.get(
        "FAIL",
        0,
    )
)


# =========================================================
# FINAL SEMANTIC FAIL IDS
# =========================================================

cell49_final_semantic_fail_results = [
    result
    for result
    in cell49_final_semantic_results
    if (
        result.get(
            "structural_status"
        )
        == "PASS"
        and
        result.get(
            "semantic_verdict"
        )
        == "FAIL"
    )
]

cell49_final_semantic_fail_candidate_ids = [
    result["candidate_id"]
    for result
    in cell49_final_semantic_fail_results
]


# =========================================================
# COST OF RECOVERY ONLY
# =========================================================

cell49_recovery_known_costs = [
    result.get(
        "estimated_cost_toman"
    )
    for result
    in cell49_recovery_results
    if isinstance(
        result.get(
            "estimated_cost_toman"
        ),
        (int, float),
    )
]

cell49_recovery_estimated_cost_toman = (
    sum(
        cell49_recovery_known_costs
    )
    if cell49_recovery_known_costs
    else None
)


# =========================================================
# RUNNER HEALTH
# =========================================================

cell49_runner_checks = {
    "cell48_population_3813": (
        len(
            cell48_full_semantic_results
        )
        == 3813
    ),

    "preserved_cell48_structural_pass_3715": (
        len(
            cell49_preserved_results
        )
        == 3715
    ),

    "recovery_population_98": (
        len(
            cell49_requests
        )
        == 98
    ),

    "source_api_failures_90": (
        len(
            cell49_api_failed_source_results
        )
        == 90
    ),

    "source_length_failures_8": (
        len(
            cell49_length_source_results
        )
        == 8
    ),

    "all_recovery_checkpoints_present": (
        cell49_valid_checkpoint_count
        == 98
    ),

    "zero_missing_recovery_checkpoints": (
        cell49_missing_checkpoint_count
        == 0
    ),

    "zero_invalid_recovery_checkpoints": (
        cell49_invalid_checkpoint_count
        == 0
    ),

    "final_population_3813": (
        len(
            cell49_final_semantic_results
        )
        == 3813
    ),

    "final_candidate_identity_exact": (
        len({
            result["candidate_id"]
            for result
            in cell49_final_semantic_results
        })
        == 3813
        and
        set(
            result["candidate_id"]
            for result
            in cell49_final_semantic_results
        )
        == set(
            cell49_cell48_by_candidate_id
        )
    ),
}


cell49_runner_status = (
    "PASS"
    if all(
        cell49_runner_checks.values()
    )
    else "FAIL"
)


if cell49_runner_status != "PASS":

    cell49_next_readiness = (
        "BLOCKED_CELL49_RUNNER_REPAIR_REQUIRED"
    )

elif cell49_final_api_failure_count > 0:

    cell49_next_readiness = (
        "READY_FOR_REMAINING_API_FAILURE_RECOVERY"
    )

elif cell49_final_structural_fail_count > 0:

    cell49_next_readiness = (
        "READY_FOR_REMAINING_STRUCTURAL_RECOVERY"
    )

elif cell49_final_semantic_fail_count > 0:

    cell49_next_readiness = (
        "READY_FOR_SEMANTIC_FAILURE_ADJUDICATION"
    )

elif (
    cell49_final_semantic_pass_count
    == 3813
):

    cell49_next_readiness = (
        "READY_FOR_GOLDEN_V2_FINAL_RECORD_CONSTRUCTION"
    )

else:

    cell49_next_readiness = (
        "READY_FOR_CELL49_RESULT_ANALYSIS"
    )


# =========================================================
# OUTPUT
# =========================================================

print()

print(
    "=== GOLDEN V2 TARGETED SEMANTIC "
    "TECHNICAL RECOVERY RESULT ==="
)

print()

print(
    "Recovery population:",
    len(
        cell49_requests
    ),
)

print(
    "Valid recovery checkpoints:",
    cell49_valid_checkpoint_count,
)

print(
    "Missing recovery checkpoints:",
    cell49_missing_checkpoint_count,
)

print(
    "Invalid recovery checkpoints:",
    cell49_invalid_checkpoint_count,
)

print()

print(
    "API candidates this execution:",
    cell49_api_candidates_this_execution,
)

print(
    "API attempts this execution:",
    cell49_api_attempts_this_execution,
)

print(
    "Recovery checkpoints created this execution:",
    cell49_checkpoints_created_this_execution,
)

print()

print(
    "Recovery structural distribution:",
    dict(
        cell49_recovery_structural_distribution
    ),
)

print(
    "Recovery execution distribution:",
    dict(
        cell49_recovery_execution_distribution
    ),
)

print(
    "Recovery finish distribution:",
    dict(
        cell49_recovery_finish_distribution
    ),
)

print(
    "Recovery semantic distribution:",
    dict(
        cell49_recovery_semantic_distribution
    ),
)

print(
    "Estimated recovery cost (toman):",
    cell49_recovery_estimated_cost_toman,
)

print()

print(
    "=== FINAL 3813 SEMANTIC STATE AFTER RECOVERY ==="
)

print(
    "Population:",
    len(
        cell49_final_semantic_results
    ),
)

print(
    "Structural distribution:",
    dict(
        cell49_final_structural_distribution
    ),
)

print(
    "Structural PASS:",
    cell49_final_structural_pass_count,
    "/ 3813",
)

print(
    "Structural FAIL:",
    cell49_final_structural_fail_count,
    "/ 3813",
)

print(
    "API failures:",
    cell49_final_api_failure_count,
)

print(
    "Semantic distribution:",
    dict(
        cell49_final_semantic_distribution
    ),
)

print(
    "Semantic PASS:",
    cell49_final_semantic_pass_count,
)

print(
    "Semantic FAIL:",
    cell49_final_semantic_fail_count,
)

print()

print(
    "Runner checks:"
)

for (
    check_name,
    check_value,
) in cell49_runner_checks.items():

    print(
        " -",
        check_name,
        ":",
        (
            "PASS"
            if check_value
            else "FAIL"
        ),
    )

print()

print(
    "Runner health:",
    cell49_runner_status,
)

print(
    "Next readiness:",
    cell49_next_readiness,
)

print(
    "CELL 49 CHECKPOINTED RUNNER:",
    cell49_runner_status,
)


assert cell49_runner_status == "PASS", (
    "Cell 49 checkpointed runner health failed."
)
````

<div style="direction: rtl; white-space: normal; line-height: 1.7;">
<h2>مرحله ۵۰ — تحلیل و تعیین تکلیف ۳۷ Semantic Failure نهایی</h2>
<p>
پس از Recovery کامل Failureهای فنی، هر ۳۸۱۳ Question/Answer اکنون دارای Judgment ساختاری معتبر هستند: ۳۷۷۶ مورد Semantic PASS و ۳۷ مورد Semantic FAIL. بنابراین دیگر هیچ API Failure، پاسخ ناقص یا مشکل ساختاری در جمعیت اعتبارسنجی وجود ندارد و این ۳۷ مورد تنها Failureهای معنایی واقعی باقی‌مانده‌اند.
</p>
<p>
در این مرحله هیچ فراخوانی جدیدی به مدل انجام نمی‌شود. برای هر یک از ۳۷ Candidate، Claim، Evidence، Question، Expected Answer، Judgment اعتبارسنج، ابعاد Failشده و Failure Reasonها کنار یکدیگر قرار می‌گیرند تا نوع خطا به‌صورت قطعی مشخص شود. هدف این مرحله تفکیک Failureهای ناشی از Question Target Drift، Answer Drift، Scope Fidelity، Modality/Quantifier Fidelity، Neighbor Fact Drift، External Knowledge و سایر ابعاد اعتبارسنجی است.
</p>
<p>
این مرحله هیچ Question/Answer را اصلاح یا حذف نمی‌کند و هیچ Judgment جدیدی تولید نمی‌کند. خروجی آن یک Repair Plan قطعی برای همان Candidateهای Failشده است تا مرحله بعد فقط موارد واقعاً نیازمند اصلاح را دوباره تولید کند و ۳۷۷۶ مورد Semantic PASS بدون تغییر حفظ شوند.
</p>
</div>

````python
# Code Cell 50 — Zero-API Adjudication and Repair Planning
# for the 37 Final Semantic Failures

from collections import Counter
import json


# =========================================================
# REQUIRED STATE
# =========================================================

CELL50_VERSION = (
    "golden_v2_semantic_failure_adjudication_v1"
)

cell50_required_names = [
    "cell25_candidate_skeletons",
    "cell47_final_generation_results",
    "cell49_final_semantic_results",
    "CELL37_BOOLEAN_FIELDS",
]

cell50_missing_state = [
    name
    for name in cell50_required_names
    if name not in globals()
]

assert not cell50_missing_state, (
    "Missing required Cell 50 state: "
    + ", ".join(cell50_missing_state)
)


# =========================================================
# EXACT LOOKUPS
# =========================================================

assert len(
    cell49_final_semantic_results
) == 3813

assert len({
    result["candidate_id"]
    for result
    in cell49_final_semantic_results
}) == 3813


cell50_skeleton_by_candidate_id = {
    record["candidate_id"]: record
    for record in cell25_candidate_skeletons
}

cell50_generation_by_candidate_id = {
    result["candidate_id"]: result
    for result in cell47_final_generation_results
}

cell50_semantic_by_candidate_id = {
    result["candidate_id"]: result
    for result in cell49_final_semantic_results
}


assert (
    set(cell50_skeleton_by_candidate_id)
    ==
    set(cell50_generation_by_candidate_id)
    ==
    set(cell50_semantic_by_candidate_id)
), (
    "Candidate identity mismatch across "
    "Cell 25, Cell 47 and Cell 49."
)


# =========================================================
# FINAL SEMANTIC STATE
# =========================================================

cell50_semantic_pass_results = [
    result
    for result
    in cell49_final_semantic_results
    if (
        result.get("structural_status") == "PASS"
        and
        result.get("semantic_verdict") == "PASS"
    )
]

cell50_semantic_fail_results = [
    result
    for result
    in cell49_final_semantic_results
    if (
        result.get("structural_status") == "PASS"
        and
        result.get("semantic_verdict") == "FAIL"
    )
]


assert len(
    cell50_semantic_pass_results
) == 3776, (
    "Expected exactly 3776 final Semantic PASS results."
)

assert len(
    cell50_semantic_fail_results
) == 37, (
    "Expected exactly 37 final Semantic FAIL results."
)


cell50_semantic_fail_candidate_ids = [
    result["candidate_id"]
    for result
    in cell50_semantic_fail_results
]

assert len(
    set(
        cell50_semantic_fail_candidate_ids
    )
) == 37


# =========================================================
# FAILURE CLASSIFICATION
# =========================================================

def cell50_failed_dimensions(validation):

    failed = []

    for dimension in CELL37_BOOLEAN_FIELDS:

        if validation.get(
            dimension
        ) is False:

            failed.append(
                dimension
            )

    return failed


def cell50_classify_failure(
    failed_dimensions,
):

    failed = set(
        failed_dimensions
    )

    classes = []

    if (
        "question_targets_claim"
        in failed
        or
        "answer_question_alignment"
        in failed
    ):

        classes.append(
            "QUESTION_TARGET_OR_ALIGNMENT_DRIFT"
        )

    if (
        "answer_recovers_claim"
        in failed
    ):

        classes.append(
            "ANSWER_CLAIM_RECOVERY_DRIFT"
        )

    if (
        "scope_fidelity"
        in failed
    ):

        classes.append(
            "SCOPE_FIDELITY_DRIFT"
        )

    if (
        "modality_quantifier_fidelity"
        in failed
    ):

        classes.append(
            "MODALITY_OR_QUANTIFIER_DRIFT"
        )

    if (
        "standalone_question"
        in failed
    ):

        classes.append(
            "QUESTION_NOT_STANDALONE"
        )

    if (
        "no_neighbor_fact_drift"
        in failed
    ):

        classes.append(
            "NEIGHBOR_FACT_DRIFT"
        )

    if (
        "no_external_knowledge"
        in failed
    ):

        classes.append(
            "EXTERNAL_KNOWLEDGE_DRIFT"
        )

    if not classes:

        classes.append(
            "UNCLASSIFIED_SEMANTIC_FAILURE"
        )

    return classes


# =========================================================
# BUILD EXACT ADJUDICATION RECORDS
# =========================================================

cell50_failure_adjudications = []


for semantic_result in (
    cell50_semantic_fail_results
):

    candidate_id = (
        semantic_result[
            "candidate_id"
        ]
    )

    skeleton = (
        cell50_skeleton_by_candidate_id[
            candidate_id
        ]
    )

    generation_result = (
        cell50_generation_by_candidate_id[
            candidate_id
        ]
    )

    generated_qa = (
        generation_result.get(
            "parsed_json"
        )
        or {}
    )

    validation = (
        semantic_result.get(
            "validation"
        )
        or {}
    )

    failed_dimensions = (
        cell50_failed_dimensions(
            validation
        )
    )

    failure_classes = (
        cell50_classify_failure(
            failed_dimensions
        )
    )

    failure_reasons = [
        str(reason).strip()
        for reason
        in (
            validation.get(
                "failure_reasons",
                []
            )
            or []
        )
        if str(reason).strip()
    ]

    adjudication = {
        "candidate_id": candidate_id,

        "claim_id": skeleton[
            "claim_id"
        ],

        "claim_text_fa": skeleton[
            "claim_text_fa"
        ],

        "evidence_ids": list(
            skeleton[
                "evidence_ids"
            ]
        ),

        "evidence_quotes": list(
            skeleton[
                "evidence_quotes"
            ]
        ),

        "question": generated_qa.get(
            "question"
        ),

        "expected_answer": (
            generated_qa.get(
                "expected_answer"
            )
        ),

        "question_intent": (
            generated_qa.get(
                "question_intent"
            )
        ),

        "failed_dimensions": (
            failed_dimensions
        ),

        "failure_classes": (
            failure_classes
        ),

        "failure_reasons": (
            failure_reasons
        ),

        # The Claim/Evidence pair has already passed
        # upstream validation. Therefore the current
        # remediation target is the generated QA.
        "repair_target": (
            "GENERATED_QA"
        ),

        "preserve_claim": True,
        "preserve_evidence": True,

        "requires_targeted_qa_regeneration": True,
    }

    cell50_failure_adjudications.append(
        adjudication
    )


assert len(
    cell50_failure_adjudications
) == 37


# =========================================================
# DISTRIBUTIONS
# =========================================================

cell50_dimension_failure_distribution = Counter()

cell50_failure_class_distribution = Counter()


for adjudication in (
    cell50_failure_adjudications
):

    for dimension in (
        adjudication[
            "failed_dimensions"
        ]
    ):

        cell50_dimension_failure_distribution[
            dimension
        ] += 1

    for failure_class in (
        adjudication[
            "failure_classes"
        ]
    ):

        cell50_failure_class_distribution[
            failure_class
        ] += 1


# =========================================================
# REPAIR POPULATION
# =========================================================

cell50_repair_candidate_ids = [
    adjudication["candidate_id"]
    for adjudication
    in cell50_failure_adjudications
    if adjudication[
        "requires_targeted_qa_regeneration"
    ]
]

assert len(
    cell50_repair_candidate_ids
) == 37

assert len(
    set(
        cell50_repair_candidate_ids
    )
) == 37


# =========================================================
# SAFETY / SCOPE CHECKS
# =========================================================

cell50_checks = {
    "full_population_3813": (
        len(
            cell49_final_semantic_results
        )
        == 3813
    ),

    "semantic_pass_3776": (
        len(
            cell50_semantic_pass_results
        )
        == 3776
    ),

    "semantic_fail_37": (
        len(
            cell50_semantic_fail_results
        )
        == 37
    ),

    "adjudication_population_37": (
        len(
            cell50_failure_adjudications
        )
        == 37
    ),

    "repair_population_37": (
        len(
            cell50_repair_candidate_ids
        )
        == 37
    ),

    "repair_ids_unique": (
        len(
            set(
                cell50_repair_candidate_ids
            )
        )
        == 37
    ),

    "all_repair_targets_generated_qa": (
        all(
            item[
                "repair_target"
            ]
            == "GENERATED_QA"
            for item
            in cell50_failure_adjudications
        )
    ),

    "claim_preserved_for_all": (
        all(
            item[
                "preserve_claim"
            ]
            is True
            for item
            in cell50_failure_adjudications
        )
    ),

    "evidence_preserved_for_all": (
        all(
            item[
                "preserve_evidence"
            ]
            is True
            for item
            in cell50_failure_adjudications
        )
    ),

    "zero_api_calls": True,
    "zero_file_writes": True,
}


cell50_status = (
    "PASS"
    if all(
        cell50_checks.values()
    )
    else "FAIL"
)


cell50_next_readiness = (
    "READY_FOR_TARGETED_QA_REPAIR"
    if cell50_status == "PASS"
    else
    "BLOCKED_CELL50_REPAIR_REQUIRED"
)


# =========================================================
# OUTPUT
# =========================================================

print(
    "=== GOLDEN V2 SEMANTIC FAILURE "
    "ADJUDICATION ==="
)

print()

print(
    "Full population:",
    len(
        cell49_final_semantic_results
    ),
)

print(
    "Semantic PASS preserved:",
    len(
        cell50_semantic_pass_results
    ),
)

print(
    "Semantic FAIL population:",
    len(
        cell50_semantic_fail_results
    ),
)

print(
    "Targeted QA repair population:",
    len(
        cell50_repair_candidate_ids
    ),
)

print()

print(
    "Failed dimension distribution:",
    dict(
        cell50_dimension_failure_distribution
    ),
)

print(
    "Failure class distribution:",
    dict(
        cell50_failure_class_distribution
    ),
)

print()


for index, adjudication in enumerate(
    cell50_failure_adjudications,
    start=1,
):

    print(
        f"--- FAILURE {index:02d}/37 ---"
    )

    print(
        "Candidate:",
        adjudication[
            "candidate_id"
        ],
    )

    print(
        "Claim:",
        adjudication[
            "claim_text_fa"
        ],
    )

    print(
        "Question:",
        adjudication[
            "question"
        ],
    )

    print(
        "Expected Answer:",
        adjudication[
            "expected_answer"
        ],
    )

    print(
        "Failed dimensions:",
        adjudication[
            "failed_dimensions"
        ],
    )

    print(
        "Failure classes:",
        adjudication[
            "failure_classes"
        ],
    )

    print(
        "Failure reasons:",
        adjudication[
            "failure_reasons"
        ],
    )

    print()


print(
    "Checks:"
)

for (
    check_name,
    check_value,
) in cell50_checks.items():

    print(
        " -",
        check_name,
        ":",
        (
            "PASS"
            if check_value
            else "FAIL"
        ),
    )

print()

print(
    "CELL 50 STATUS:",
    cell50_status,
)

print(
    "Next readiness:",
    cell50_next_readiness,
)


assert cell50_status == "PASS", (
    "Cell 50 adjudication failed."
)
````

<p>
در مرحله قبل، هر ۳۷ Semantic Failure نهایی به‌صورت قطعی تحلیل شد و مشخص گردید که هدف Repair در تمام موارد، Generated Question/Answer است و Claim و Evidence معتبر قبلی باید بدون تغییر حفظ شوند. ۳۷۷۶ Question/Answer دارای Semantic PASS نیز در این مرحله کاملاً محفوظ می‌مانند و دوباره به مدل ارسال نمی‌شوند.
</p>
<p>
در این مرحله فقط ۳۷ Candidate ناموفق با Generation Protocol v4 بازتولید می‌شوند. علاوه بر Claim و Evidence، Failure Dimensionها و Failure Reasonهای واقعی هر Candidate نیز در Prompt قرار می‌گیرند تا مدل دقیقاً خطای قبلی را اصلاح کند؛ از جمله Question Target Drift، Answer Alignment، Scope Fidelity، Modality و Quantifier، Neighbor Fact Drift، External Knowledge و Standalone بودن Question.
</p>
<p>
اجرای این مرحله Checkpoint-first، Fingerprint-bound و قابل Resume است. هر Candidate دارای Recovery Checkpoint مستقل خواهد بود و اجرای مجدد فقط Candidate فاقد Checkpoint معتبر را به مدل ارسال می‌کند. این مرحله فقط تولید QA اصلاح‌شده را انجام می‌دهد و پذیرش معنایی آن‌ها در مرحله بعد با Validator مستقل انجام خواهد شد.
</p>

````python
# Code Cell 51 — Targeted QA Regeneration for 37 Semantic Failures

from collections import Counter
from concurrent.futures import ThreadPoolExecutor, as_completed
from pathlib import Path
import hashlib
import json
import os
import re
import tempfile
import threading
import time


# =========================================================
# CONFIG
# =========================================================

CELL51_RUNNER_VERSION = (
    "golden_v2_targeted_qa_semantic_repair_runner_v1"
)

CELL51_GENERATION_PROTOCOL_VERSION = (
    GENERATION_V4_VERSION
)

CELL51_PROMPT_VERSION = (
    "golden_v2_targeted_semantic_repair_prompt_v1"
)

CELL51_OUTPUT_SCHEMA_VERSION = (
    GENERATION_V4_OUTPUT_SCHEMA_VERSION
)

CELL51_MAX_TOKENS = 1536
CELL51_TEMPERATURE = 0.7
CELL51_TOP_P = 0.8

CELL51_MAX_ATTEMPTS = 2
CELL51_RETRY_SECONDS = 2.0

CELL51_MAX_WORKERS = 16
CELL51_REQUEST_START_SPACING_SECONDS = 0.6
CELL51_REQUEST_TIMEOUT_SECONDS = 180.0

# Paid execution is allowed only for missing exact repair checkpoints.
# Before final Restart -> Run All this MUST become False.
CELL51_ALLOW_API_FOR_MISSING = False


# =========================================================
# REQUIRED STATE
# =========================================================

cell51_required_names = [
    "project_root",
    "semantic_review_client",
    "semantic_review_model",
    "cell25_candidate_skeletons",
    "cell47_final_generation_results",
    "cell50_failure_adjudications",
    "cell50_repair_candidate_ids",
    "GOLDEN_V2_GENERATION_SYSTEM_PROMPT_V4",
    "cell41_validate_generation_output_v4",
]

cell51_missing_state = [
    name
    for name in cell51_required_names
    if name not in globals()
]

assert not cell51_missing_state, (
    "Missing required Cell 51 state: "
    + ", ".join(cell51_missing_state)
)


# =========================================================
# EXACT POPULATION
# =========================================================

assert len(
    cell50_repair_candidate_ids
) == 37

assert len(
    set(
        cell50_repair_candidate_ids
    )
) == 37

assert len(
    cell50_failure_adjudications
) == 37


cell51_skeleton_by_candidate_id = {
    record["candidate_id"]: record
    for record in cell25_candidate_skeletons
}

cell51_previous_generation_by_candidate_id = {
    result["candidate_id"]: result
    for result in cell47_final_generation_results
}

cell51_adjudication_by_candidate_id = {
    record["candidate_id"]: record
    for record in cell50_failure_adjudications
}


assert (
    set(
        cell50_repair_candidate_ids
    )
    ==
    set(
        cell51_adjudication_by_candidate_id
    )
)

assert set(
    cell50_repair_candidate_ids
).issubset(
    cell51_skeleton_by_candidate_id
)

assert set(
    cell50_repair_candidate_ids
).issubset(
    cell51_previous_generation_by_candidate_id
)


# =========================================================
# HASH UTILITIES
# =========================================================

def cell51_canonical_json(value):

    return json.dumps(
        value,
        ensure_ascii=False,
        sort_keys=True,
        separators=(",", ":"),
    )


def cell51_sha256_text(text):

    return hashlib.sha256(
        text.encode("utf-8")
    ).hexdigest()


# =========================================================
# REPAIR PACKET
# =========================================================

def cell51_build_repair_packet(
    candidate_id,
):

    skeleton = (
        cell51_skeleton_by_candidate_id[
            candidate_id
        ]
    )

    previous_generation = (
        cell51_previous_generation_by_candidate_id[
            candidate_id
        ]
    )

    adjudication = (
        cell51_adjudication_by_candidate_id[
            candidate_id
        ]
    )

    previous_qa = (
        previous_generation.get(
            "parsed_json"
        )
        or {}
    )

    evidence_ids = list(
        skeleton[
            "evidence_ids"
        ]
    )

    evidence_quotes = list(
        skeleton[
            "evidence_quotes"
        ]
    )

    assert len(
        evidence_ids
    ) == len(
        evidence_quotes
    )

    evidence = []

    for index, (
        evidence_id,
        evidence_quote,
    ) in enumerate(
        zip(
            evidence_ids,
            evidence_quotes,
        ),
        start=1,
    ):

        evidence.append({
            "index": index,
            "evidence_id": evidence_id,
            "quote": evidence_quote,
        })

    return {
        "candidate_id": candidate_id,

        "claim": {
            "claim_id": (
                skeleton[
                    "claim_id"
                ]
            ),

            "claim_text_fa": (
                skeleton[
                    "claim_text_fa"
                ]
            ),

            "claim_type": (
                skeleton.get(
                    "claim_type"
                )
            ),

            "claim_types": (
                skeleton.get(
                    "claim_types",
                    [],
                )
            ),

            "authority_class": (
                skeleton.get(
                    "authority_class"
                )
            ),
        },

        "evidence": evidence,

        "previous_generated_qa": {
            "question": (
                previous_qa.get(
                    "question"
                )
            ),

            "expected_answer": (
                previous_qa.get(
                    "expected_answer"
                )
            ),

            "question_intent": (
                previous_qa.get(
                    "question_intent"
                )
            ),
        },

        "semantic_failure": {
            "failed_dimensions": list(
                adjudication[
                    "failed_dimensions"
                ]
            ),

            "failure_classes": list(
                adjudication[
                    "failure_classes"
                ]
            ),

            "failure_reasons": list(
                adjudication[
                    "failure_reasons"
                ]
            ),
        },
    }


# =========================================================
# TARGETED REPAIR PROMPT
# =========================================================

def cell51_build_user_prompt(
    candidate_id,
):

    packet = (
        cell51_build_repair_packet(
            candidate_id
        )
    )

    return (
        "Generate a corrected Question/Answer pair for the supplied "
        "prevalidated Claim and Evidence.\n\n"

        "This is a targeted repair of a previously generated QA pair "
        "that failed independent semantic validation.\n\n"

        "The Claim and Evidence are authoritative for this task. "
        "Do NOT repair, rewrite, expand, reinterpret, reject or "
        "revalidate the Claim or Evidence.\n\n"

        "Use the previous QA and the supplied semantic failure "
        "dimensions/reasons only to understand what must be fixed. "
        "Do NOT preserve wording from the previous QA when that "
        "wording caused the failure.\n\n"

        "REPAIR REQUIREMENTS:\n"
        "1. The Question must directly target information actually "
        "asserted by the Claim.\n"
        "2. The Expected Answer must directly answer that Question "
        "and faithfully recover the relevant Claim meaning.\n"
        "3. Preserve all meaning-bearing scope restrictions.\n"
        "4. Preserve negation and modality such as can/may/should/"
        "must/possible/necessary and their Persian equivalents.\n"
        "5. Preserve quantifiers and numerical qualifiers such as "
        "only/تنها/فقط, all, some, many, most, at least, at most, "
        "exactly, percentages, quantities and comparisons whenever "
        "they are part of the target meaning.\n"
        "6. Never ask WHY, HOW, cause, reason, mechanism or consequence "
        "unless the Claim explicitly asserts that information.\n"
        "7. Never ask for a number, amount, identity, location or "
        "specific detail that the Claim does not provide.\n"
        "8. Do not introduce facts from neighboring Evidence that are "
        "not part of the target Claim.\n"
        "9. Do not introduce external knowledge, inferred units, "
        "currencies, entities or explanations.\n"
        "10. The Question must be standalone and unambiguous.\n"
        "11. Prefer a simple direct Question over a clever or "
        "linguistically complex Question.\n"
        "12. Repair every supplied failed dimension; do not merely "
        "paraphrase the previous failed QA.\n\n"

        "Return ONLY one JSON object with exactly these fields:\n"
        "{\n"
        '  "question": "...",\n'
        '  "expected_answer": "...",\n'
        '  "question_intent": "..."\n'
        "}\n\n"

        "REPAIR PACKET:\n"
        + json.dumps(
            packet,
            ensure_ascii=False,
            sort_keys=True,
            indent=2,
        )
    )


# =========================================================
# REQUEST CONTRACT
# =========================================================

def cell51_build_request(
    candidate_id,
):

    repair_packet = (
        cell51_build_repair_packet(
            candidate_id
        )
    )

    user_prompt = (
        cell51_build_user_prompt(
            candidate_id
        )
    )

    previous_qa = (
        cell51_previous_generation_by_candidate_id[
            candidate_id
        ].get(
            "parsed_json"
        )
        or {}
    )

    contract = {
        "runner_version": (
            CELL51_RUNNER_VERSION
        ),

        "generation_protocol_version": (
            CELL51_GENERATION_PROTOCOL_VERSION
        ),

        "prompt_version": (
            CELL51_PROMPT_VERSION
        ),

        "output_schema_version": (
            CELL51_OUTPUT_SCHEMA_VERSION
        ),

        "candidate_id": (
            candidate_id
        ),

        "repair_packet_sha256": (
            cell51_sha256_text(
                cell51_canonical_json(
                    repair_packet
                )
            )
        ),

        "previous_qa_sha256": (
            cell51_sha256_text(
                cell51_canonical_json(
                    previous_qa
                )
            )
        ),

        "model": (
            semantic_review_model
        ),

        "system_prompt": (
            GOLDEN_V2_GENERATION_SYSTEM_PROMPT_V4
        ),

        "user_prompt": (
            user_prompt
        ),

        "temperature": (
            CELL51_TEMPERATURE
        ),

        "top_p": (
            CELL51_TOP_P
        ),

        "max_tokens": (
            CELL51_MAX_TOKENS
        ),

        "enable_thinking": False,
    }

    request_fingerprint = (
        cell51_sha256_text(
            cell51_canonical_json(
                contract
            )
        )
    )

    return {
        **contract,

        "request_fingerprint": (
            request_fingerprint
        ),
    }


cell51_requests = [
    cell51_build_request(
        candidate_id
    )
    for candidate_id
    in cell50_repair_candidate_ids
]


assert len(
    cell51_requests
) == 37

assert len({
    request["candidate_id"]
    for request
    in cell51_requests
}) == 37

assert len({
    request["request_fingerprint"]
    for request
    in cell51_requests
}) == 37


# =========================================================
# CHECKPOINT NAMESPACE
# =========================================================

cell51_checkpoint_dir = (
    Path(project_root)
    / "artifacts"
    / "phase2"
    / "golden_dataset_v2"
    / "work"
    / "targeted_qa_semantic_repair_v1"
)

cell51_checkpoint_dir.mkdir(
    parents=True,
    exist_ok=True,
)


def cell51_checkpoint_path(
    candidate_id,
):

    safe_id = re.sub(
        r"[^A-Za-z0-9_.-]+",
        "_",
        candidate_id,
    )

    return (
        cell51_checkpoint_dir
        / f"{safe_id}.json"
    )


# =========================================================
# ATOMIC WRITE
# =========================================================

def cell51_atomic_write_json(
    path,
    payload,
):

    path = Path(path)

    content = (
        json.dumps(
            payload,
            ensure_ascii=False,
            sort_keys=True,
            indent=2,
        )
        + "\n"
    )

    fd, temp_name = tempfile.mkstemp(
        prefix=path.name + ".",
        suffix=".tmp",
        dir=str(path.parent),
    )

    try:

        with os.fdopen(
            fd,
            "w",
            encoding="utf-8",
        ) as handle:

            handle.write(
                content
            )

            handle.flush()

            os.fsync(
                handle.fileno()
            )

        os.replace(
            temp_name,
            path,
        )

    except Exception:

        try:

            os.unlink(
                temp_name
            )

        except FileNotFoundError:

            pass

        raise


# =========================================================
# CHECKPOINT CONTRACT
# =========================================================

def cell51_build_checkpoint(
    request,
    result,
):

    checkpoint = {
        "checkpoint_version": (
            CELL51_RUNNER_VERSION
        ),

        "candidate_id": (
            request[
                "candidate_id"
            ]
        ),

        "request_fingerprint": (
            request[
                "request_fingerprint"
            ]
        ),

        "repair_packet_sha256": (
            request[
                "repair_packet_sha256"
            ]
        ),

        "previous_qa_sha256": (
            request[
                "previous_qa_sha256"
            ]
        ),

        "generation_protocol_version": (
            CELL51_GENERATION_PROTOCOL_VERSION
        ),

        "prompt_version": (
            CELL51_PROMPT_VERSION
        ),

        "output_schema_version": (
            CELL51_OUTPUT_SCHEMA_VERSION
        ),

        "model": (
            semantic_review_model
        ),

        "temperature": (
            CELL51_TEMPERATURE
        ),

        "top_p": (
            CELL51_TOP_P
        ),

        "max_tokens": (
            CELL51_MAX_TOKENS
        ),

        "result": (
            result
        ),
    }

    checkpoint[
        "checkpoint_payload_sha256"
    ] = cell51_sha256_text(
        cell51_canonical_json(
            checkpoint
        )
    )

    return checkpoint


def cell51_validate_checkpoint(
    checkpoint,
    request,
):

    errors = []

    if not isinstance(
        checkpoint,
        dict,
    ):

        return [
            "CHECKPOINT_NOT_OBJECT"
        ]

    stored_sha = checkpoint.get(
        "checkpoint_payload_sha256"
    )

    payload_without_sha = {
        key: value
        for key, value
        in checkpoint.items()
        if key
        != "checkpoint_payload_sha256"
    }

    recalculated_sha = (
        cell51_sha256_text(
            cell51_canonical_json(
                payload_without_sha
            )
        )
    )

    if stored_sha != recalculated_sha:

        errors.append(
            "CHECKPOINT_HASH_MISMATCH"
        )

    expected_values = {
        "checkpoint_version": (
            CELL51_RUNNER_VERSION
        ),

        "candidate_id": (
            request[
                "candidate_id"
            ]
        ),

        "request_fingerprint": (
            request[
                "request_fingerprint"
            ]
        ),

        "repair_packet_sha256": (
            request[
                "repair_packet_sha256"
            ]
        ),

        "previous_qa_sha256": (
            request[
                "previous_qa_sha256"
            ]
        ),

        "generation_protocol_version": (
            CELL51_GENERATION_PROTOCOL_VERSION
        ),

        "prompt_version": (
            CELL51_PROMPT_VERSION
        ),

        "output_schema_version": (
            CELL51_OUTPUT_SCHEMA_VERSION
        ),

        "model": (
            semantic_review_model
        ),

        "temperature": (
            CELL51_TEMPERATURE
        ),

        "top_p": (
            CELL51_TOP_P
        ),

        "max_tokens": (
            CELL51_MAX_TOKENS
        ),
    }

    for (
        field_name,
        expected_value,
    ) in expected_values.items():

        if (
            checkpoint.get(
                field_name
            )
            != expected_value
        ):

            errors.append(
                f"{field_name.upper()}_MISMATCH"
            )

    result = checkpoint.get(
        "result"
    )

    if not isinstance(
        result,
        dict,
    ):

        errors.append(
            "CHECKPOINT_RESULT_INVALID"
        )

    elif (
        result.get(
            "candidate_id"
        )
        != request[
            "candidate_id"
        ]
    ):

        errors.append(
            "CHECKPOINT_RESULT_CANDIDATE_MISMATCH"
        )

    return sorted(
        set(
            errors
        )
    )


def cell51_load_checkpoint(
    request,
):

    path = (
        cell51_checkpoint_path(
            request[
                "candidate_id"
            ]
        )
    )

    if not path.exists():

        return (
            None,
            "MISSING",
            [],
        )

    try:

        with path.open(
            "r",
            encoding="utf-8",
        ) as handle:

            checkpoint = (
                json.load(
                    handle
                )
            )

    except Exception as exc:

        return (
            None,
            "UNREADABLE",
            [
                f"{type(exc).__name__}: {exc}"
            ],
        )

    errors = (
        cell51_validate_checkpoint(
            checkpoint,
            request,
        )
    )

    if errors:

        return (
            checkpoint,
            "INVALID",
            errors,
        )

    return (
        checkpoint,
        "VALID",
        [],
    )


# =========================================================
# RESPONSE UTILITIES
# =========================================================

def cell51_clean_model_json(
    content,
):

    if not isinstance(
        content,
        str,
    ):

        return ""

    clean = (
        content.strip()
    )

    if clean.startswith(
        "```"
    ):

        clean = re.sub(
            r"^```(?:json)?\s*",
            "",
            clean,
            flags=re.IGNORECASE,
        )

        clean = re.sub(
            r"\s*```$",
            "",
            clean,
        )

    return clean.strip()


def cell51_is_rate_limit_exception(
    exc,
):

    error_text = str(
        exc
    ).lower()

    error_type = type(
        exc
    ).__name__.lower()

    return (
        "ratelimit"
        in error_type
        or
        "rate_limit"
        in error_type
        or
        "429"
        in error_text
        or
        "too many requests"
        in error_text
    )


# =========================================================
# REQUEST THROTTLE
# =========================================================

cell51_request_start_lock = (
    threading.Lock()
)

cell51_last_request_start = [
    0.0
]


def cell51_wait_for_request_slot():

    with cell51_request_start_lock:

        now = (
            time.monotonic()
        )

        elapsed = (
            now
            - cell51_last_request_start[
                0
            ]
        )

        wait_seconds = max(
            0.0,
            CELL51_REQUEST_START_SPACING_SECONDS
            - elapsed,
        )

        if wait_seconds > 0:

            time.sleep(
                wait_seconds
            )

        cell51_last_request_start[
            0
        ] = time.monotonic()


# =========================================================
# COST
# =========================================================

cell51_has_cost_rates = (
    "INPUT_TOMAN_PER_TOKEN"
    in globals()
    and
    "OUTPUT_TOMAN_PER_TOKEN"
    in globals()
)


def cell51_estimated_cost_toman(
    input_tokens,
    output_tokens,
):

    if not cell51_has_cost_rates:

        return None

    return (
        input_tokens
        * INPUT_TOMAN_PER_TOKEN
        +
        output_tokens
        * OUTPUT_TOMAN_PER_TOKEN
    )


# =========================================================
# API FAILURE RESULT
# =========================================================

def cell51_build_api_failure_result(
    request,
    api_attempts,
    attempt_history,
):

    return {
        "candidate_id": (
            request[
                "candidate_id"
            ]
        ),

        "request_fingerprint": (
            request[
                "request_fingerprint"
            ]
        ),

        "execution_status": (
            "API_FAILED"
        ),

        "api_attempts": (
            api_attempts
        ),

        "attempt_history": (
            attempt_history
        ),

        "finish_reason": None,

        "usage": {
            "input_tokens": 0,
            "output_tokens": 0,
            "reasoning_tokens": 0,
            "total_tokens": 0,
        },

        "estimated_cost_toman": None,

        "raw_content": None,

        "parsed_json": None,

        "parse_error": None,

        "structural_errors": [
            "API_FAILED"
        ],

        "structural_status": (
            "FAIL"
        ),
    }


# =========================================================
# EXECUTION
# =========================================================

def cell51_execute_request(
    request,
):

    attempt_history = []
    api_attempts = 0

    for attempt_number in range(
        1,
        CELL51_MAX_ATTEMPTS + 1,
    ):

        cell51_wait_for_request_slot()

        api_attempts += 1

        try:

            response = (
                semantic_review_client
                .chat
                .completions
                .create(
                    model=(
                        semantic_review_model
                    ),

                    messages=[
                        {
                            "role": "system",
                            "content": (
                                GOLDEN_V2_GENERATION_SYSTEM_PROMPT_V4
                            ),
                        },
                        {
                            "role": "user",
                            "content": (
                                request[
                                    "user_prompt"
                                ]
                            ),
                        },
                    ],

                    temperature=(
                        CELL51_TEMPERATURE
                    ),

                    top_p=(
                        CELL51_TOP_P
                    ),

                    max_tokens=(
                        CELL51_MAX_TOKENS
                    ),

                    timeout=(
                        CELL51_REQUEST_TIMEOUT_SECONDS
                    ),

                    extra_body={
                        "enable_thinking": False,
                    },
                )
            )

            if response is None:

                raise RuntimeError(
                    "EMPTY_API_RESPONSE"
                )

            choices = getattr(
                response,
                "choices",
                None,
            )

            if (
                not isinstance(
                    choices,
                    (list, tuple),
                )
                or
                len(choices) == 0
            ):

                raise RuntimeError(
                    "API_RESPONSE_CHOICES_MISSING_OR_EMPTY"
                )

            choice = choices[0]

            if choice is None:

                raise RuntimeError(
                    "API_RESPONSE_FIRST_CHOICE_IS_NONE"
                )

            message = getattr(
                choice,
                "message",
                None,
            )

            if message is None:

                raise RuntimeError(
                    "API_RESPONSE_MESSAGE_MISSING"
                )

        except Exception as exc:

            attempt_history.append({
                "attempt": (
                    attempt_number
                ),

                "status": (
                    "API_ERROR"
                ),

                "error": (
                    f"{type(exc).__name__}: {exc}"
                ),
            })

            if (
                attempt_number
                < CELL51_MAX_ATTEMPTS
            ):

                wait_seconds = (
                    70.0
                    if cell51_is_rate_limit_exception(
                        exc
                    )
                    else (
                        CELL51_RETRY_SECONDS
                        * attempt_number
                    )
                )

                time.sleep(
                    wait_seconds
                )

                continue

            return (
                cell51_build_api_failure_result(
                    request,
                    api_attempts,
                    attempt_history,
                )
            )

        finish_reason = getattr(
            choice,
            "finish_reason",
            None,
        )

        raw_content = getattr(
            message,
            "content",
            None,
        )

        usage = getattr(
            response,
            "usage",
            None,
        )

        input_tokens = int(
            getattr(
                usage,
                "prompt_tokens",
                0,
            )
            or 0
        )

        output_tokens = int(
            getattr(
                usage,
                "completion_tokens",
                0,
            )
            or 0
        )

        total_tokens = int(
            getattr(
                usage,
                "total_tokens",
                0,
            )
            or (
                input_tokens
                + output_tokens
            )
        )

        reasoning_tokens = 0

        completion_details = getattr(
            usage,
            "completion_tokens_details",
            None,
        )

        if completion_details is not None:

            reasoning_tokens = int(
                getattr(
                    completion_details,
                    "reasoning_tokens",
                    0,
                )
                or 0
            )

        clean_content = (
            cell51_clean_model_json(
                raw_content
            )
        )

        parsed_json = None
        parse_error = None

        try:

            parsed_json = (
                json.loads(
                    clean_content
                )
            )

        except Exception as exc:

            parse_error = (
                f"{type(exc).__name__}: {exc}"
            )

        if parsed_json is None:

            structural_errors = [
                "JSON_PARSE_FAILED"
            ]

        else:

            structural_errors = (
                cell41_validate_generation_output_v4(
                    parsed_json
                )
            )

        structural_status = (
            "PASS"
            if not structural_errors
            else "FAIL"
        )

        return {
            "candidate_id": (
                request[
                    "candidate_id"
                ]
            ),

            "request_fingerprint": (
                request[
                    "request_fingerprint"
                ]
            ),

            "execution_status": (
                "COMPLETED"
            ),

            "api_attempts": (
                api_attempts
            ),

            "attempt_history": (
                attempt_history
            ),

            "finish_reason": (
                finish_reason
            ),

            "usage": {
                "input_tokens": (
                    input_tokens
                ),

                "output_tokens": (
                    output_tokens
                ),

                "reasoning_tokens": (
                    reasoning_tokens
                ),

                "total_tokens": (
                    total_tokens
                ),
            },

            "estimated_cost_toman": (
                cell51_estimated_cost_toman(
                    input_tokens,
                    output_tokens,
                )
            ),

            "raw_content": (
                raw_content
            ),

            "parsed_json": (
                parsed_json
            ),

            "parse_error": (
                parse_error
            ),

            "structural_errors": (
                structural_errors
            ),

            "structural_status": (
                structural_status
            ),
        }

    raise AssertionError(
        "Unexpected Cell 51 execution path."
    )


# =========================================================
# PREFLIGHT CHECKPOINT SCAN
# =========================================================

cell51_valid_before_api = {}
cell51_missing_requests = []
cell51_invalid_before_api = []


for request in cell51_requests:

    (
        checkpoint,
        status,
        errors,
    ) = cell51_load_checkpoint(
        request
    )

    candidate_id = (
        request[
            "candidate_id"
        ]
    )

    if status == "VALID":

        cell51_valid_before_api[
            candidate_id
        ] = checkpoint

    elif status == "MISSING":

        cell51_missing_requests.append(
            request
        )

    else:

        cell51_invalid_before_api.append({
            "candidate_id": (
                candidate_id
            ),

            "status": (
                status
            ),

            "errors": (
                errors
            ),
        })


assert not cell51_invalid_before_api, (
    "Invalid/unreadable Cell 51 checkpoints detected. "
    "Refusing automatic paid replacement. "
    f"Examples: {cell51_invalid_before_api[:5]}"
)


# =========================================================
# PREFLIGHT OUTPUT
# =========================================================

print(
    "=== GOLDEN V2 TARGETED QA "
    "SEMANTIC REPAIR PREFLIGHT ==="
)

print()

print(
    "Full population:",
    3813,
)

print(
    "Semantic PASS preserved:",
    3776,
)

print(
    "Targeted repair population:",
    len(
        cell51_requests
    ),
)

print(
    "Exact valid repair checkpoints:",
    len(
        cell51_valid_before_api
    ),
)

print(
    "Missing repair checkpoints:",
    len(
        cell51_missing_requests
    ),
)

print(
    "Model:",
    semantic_review_model,
)

print(
    "Generation protocol:",
    CELL51_GENERATION_PROTOCOL_VERSION,
)

print(
    "Repair prompt version:",
    CELL51_PROMPT_VERSION,
)

print(
    "Max tokens:",
    CELL51_MAX_TOKENS,
)

print(
    "Parallel workers:",
    CELL51_MAX_WORKERS,
)

print(
    "Request spacing seconds:",
    CELL51_REQUEST_START_SPACING_SECONDS,
)

print(
    "Request timeout seconds:",
    CELL51_REQUEST_TIMEOUT_SECONDS,
)

print(
    "API allowed for missing exact checkpoints:",
    CELL51_ALLOW_API_FOR_MISSING,
)

print(
    "Previously accepted QA sent to API:",
    0,
)

print()


if (
    cell51_missing_requests
    and
    not CELL51_ALLOW_API_FOR_MISSING
):

    raise AssertionError(
        "Cell 51 has missing repair checkpoints "
        "but API execution is disabled."
    )


# =========================================================
# EXECUTE + CHECKPOINT
# =========================================================

cell51_api_candidates_this_execution = 0
cell51_api_attempts_this_execution = 0
cell51_checkpoints_created_this_execution = 0


def cell51_execute_and_checkpoint(
    request,
):

    candidate_id = (
        request[
            "candidate_id"
        ]
    )

    result = (
        cell51_execute_request(
            request
        )
    )

    checkpoint = (
        cell51_build_checkpoint(
            request,
            result,
        )
    )

    cell51_atomic_write_json(
        cell51_checkpoint_path(
            candidate_id
        ),
        checkpoint,
    )

    (
        verified_checkpoint,
        verified_status,
        verified_errors,
    ) = cell51_load_checkpoint(
        request
    )

    if verified_status != "VALID":

        raise AssertionError(
            "Cell 51 checkpoint verification "
            f"failed for {candidate_id}: "
            f"{verified_errors}"
        )

    return (
        candidate_id,
        verified_checkpoint[
            "result"
        ],
    )


if cell51_missing_requests:

    completed_count = 0

    executor = ThreadPoolExecutor(
        max_workers=(
            CELL51_MAX_WORKERS
        )
    )

    future_to_request = {}

    try:

        future_to_request = {
            executor.submit(
                cell51_execute_and_checkpoint,
                request,
            ): request

            for request
            in cell51_missing_requests
        }

        for future in as_completed(
            future_to_request
        ):

            request = (
                future_to_request[
                    future
                ]
            )

            candidate_id = (
                request[
                    "candidate_id"
                ]
            )

            try:

                (
                    resolved_candidate_id,
                    result,
                ) = future.result()

            except Exception as exc:

                completed_count += 1

                print(
                    f"[{completed_count:02d}/"
                    f"{len(cell51_missing_requests):02d}]",
                    candidate_id,
                    "| WORKER_EXCEPTION =",
                    f"{type(exc).__name__}: {exc}",
                    flush=True,
                )

                continue

            assert (
                resolved_candidate_id
                == candidate_id
            )

            completed_count += 1

            cell51_api_candidates_this_execution += 1

            cell51_api_attempts_this_execution += int(
                result.get(
                    "api_attempts",
                    0,
                )
                or 0
            )

            cell51_checkpoints_created_this_execution += 1

            print(
                f"[{completed_count:02d}/"
                f"{len(cell51_missing_requests):02d}]",
                candidate_id,
                "| structural=",
                result.get(
                    "structural_status"
                ),
                "| finish=",
                result.get(
                    "finish_reason"
                ),
                flush=True,
            )

    except KeyboardInterrupt:

        print(
            "\nCell 51 interrupted by user. "
            "Completed repair checkpoints remain preserved.",
            flush=True,
        )

        for future in future_to_request:

            if not future.done():

                future.cancel()

        raise

    finally:

        executor.shutdown(
            wait=False,
            cancel_futures=True,
        )


# =========================================================
# FINAL CHECKPOINT RESCAN
# =========================================================

cell51_repair_results = []

cell51_valid_checkpoint_count = 0
cell51_missing_checkpoint_count = 0
cell51_invalid_checkpoint_count = 0


for request in cell51_requests:

    (
        checkpoint,
        status,
        errors,
    ) = cell51_load_checkpoint(
        request
    )

    if status == "VALID":

        cell51_valid_checkpoint_count += 1

        cell51_repair_results.append(
            checkpoint[
                "result"
            ]
        )

    elif status == "MISSING":

        cell51_missing_checkpoint_count += 1

    else:

        cell51_invalid_checkpoint_count += 1


# =========================================================
# RESULT DISTRIBUTIONS
# =========================================================

cell51_structural_distribution = Counter(
    result.get(
        "structural_status"
    )
    for result
    in cell51_repair_results
)

cell51_execution_distribution = Counter(
    result.get(
        "execution_status"
    )
    for result
    in cell51_repair_results
)

cell51_finish_distribution = Counter(
    result.get(
        "finish_reason"
    )
    for result
    in cell51_repair_results
)


cell51_structural_pass_count = (
    cell51_structural_distribution.get(
        "PASS",
        0,
    )
)

cell51_structural_fail_count = (
    cell51_structural_distribution.get(
        "FAIL",
        0,
    )
)

cell51_api_failure_count = (
    cell51_execution_distribution.get(
        "API_FAILED",
        0,
    )
)


# =========================================================
# COST
# =========================================================

cell51_known_costs = [
    result.get(
        "estimated_cost_toman"
    )
    for result
    in cell51_repair_results
    if isinstance(
        result.get(
            "estimated_cost_toman"
        ),
        (int, float),
    )
]

cell51_estimated_cost_toman = (
    sum(
        cell51_known_costs
    )
    if cell51_known_costs
    else None
)


# =========================================================
# RUNNER HEALTH
# =========================================================

cell51_checks = {
    "repair_population_37": (
        len(
            cell51_requests
        )
        == 37
    ),

    "all_repair_checkpoints_present": (
        cell51_valid_checkpoint_count
        == 37
    ),

    "zero_missing_checkpoints": (
        cell51_missing_checkpoint_count
        == 0
    ),

    "zero_invalid_checkpoints": (
        cell51_invalid_checkpoint_count
        == 0
    ),

    "repair_result_population_37": (
        len(
            cell51_repair_results
        )
        == 37
    ),

    "repair_candidate_identity_exact": (
        set(
            result["candidate_id"]
            for result
            in cell51_repair_results
        )
        ==
        set(
            cell50_repair_candidate_ids
        )
    ),

    "previously_accepted_population_not_regenerated": True,
}


cell51_runner_status = (
    "PASS"
    if all(
        cell51_checks.values()
    )
    else "FAIL"
)


if cell51_runner_status != "PASS":

    cell51_next_readiness = (
        "BLOCKED_CELL51_RUNNER_REPAIR_REQUIRED"
    )

elif cell51_api_failure_count > 0:

    cell51_next_readiness = (
        "READY_FOR_TARGETED_QA_REPAIR_API_RECOVERY"
    )

elif cell51_structural_fail_count > 0:

    cell51_next_readiness = (
        "READY_FOR_TARGETED_QA_REPAIR_STRUCTURAL_RECOVERY"
    )

elif cell51_structural_pass_count == 37:

    cell51_next_readiness = (
        "READY_FOR_TARGETED_QA_REPAIR_SEMANTIC_REVALIDATION"
    )

else:

    cell51_next_readiness = (
        "READY_FOR_CELL51_RESULT_ANALYSIS"
    )


# =========================================================
# OUTPUT
# =========================================================

print()

print(
    "=== GOLDEN V2 TARGETED QA "
    "SEMANTIC REPAIR RESULT ==="
)

print()

print(
    "Repair population:",
    len(
        cell51_requests
    ),
)

print(
    "Valid repair checkpoints:",
    cell51_valid_checkpoint_count,
)

print(
    "Missing repair checkpoints:",
    cell51_missing_checkpoint_count,
)

print(
    "Invalid repair checkpoints:",
    cell51_invalid_checkpoint_count,
)

print()

print(
    "API candidates this execution:",
    cell51_api_candidates_this_execution,
)

print(
    "API attempts this execution:",
    cell51_api_attempts_this_execution,
)

print(
    "Repair checkpoints created this execution:",
    cell51_checkpoints_created_this_execution,
)

print()

print(
    "Structural distribution:",
    dict(
        cell51_structural_distribution
    ),
)

print(
    "Structural PASS:",
    cell51_structural_pass_count,
    "/ 37",
)

print(
    "Structural FAIL:",
    cell51_structural_fail_count,
    "/ 37",
)

print(
    "API failures:",
    cell51_api_failure_count,
)

print(
    "Finish reason distribution:",
    dict(
        cell51_finish_distribution
    ),
)

print(
    "Estimated repair cost (toman):",
    cell51_estimated_cost_toman,
)

print()

print(
    "Runner checks:"
)

for (
    check_name,
    check_value,
) in cell51_checks.items():

    print(
        " -",
        check_name,
        ":",
        (
            "PASS"
            if check_value
            else "FAIL"
        ),
    )

print()

print(
    "Runner health:",
    cell51_runner_status,
)

print(
    "Next readiness:",
    cell51_next_readiness,
)

print(
    "CELL 51 CHECKPOINTED RUNNER:",
    cell51_runner_status,
)


assert cell51_runner_status == "PASS", (
    "Cell 51 checkpointed runner health failed."
)
````

در مرحله قبل، بازتولید هدفمند ۳۷ Question/Answer دارای Semantic Failure انجام شد. از این جمعیت، ۳۰ خروجی با ساختار معتبر تولید شدند و ۷ خروجی صرفاً به دلیل پایان پاسخ با finish_reason=length و ناقص ماندن JSON از نظر ساختاری Fail شدند. هیچ API Failure در این مرحله باقی نمانده است.
</p>
<p>
در این مرحله فقط همان ۷ Candidate دارای Failure ساختاری دوباره تولید می‌شوند. ۳۰ Repair ساختاری سالم مرحله قبل بدون تغییر حفظ می‌شوند و ۳۷۷۶ Question/Answer دارای Semantic PASS نیز به هیچ عنوان دوباره به مدل ارسال نخواهند شد. برای کاهش احتمال تکرار Truncation، سقف خروجی از ۱۵۳۶ به ۳۰۷۲ Token افزایش می‌یابد، در حالی که Generation Protocol، Repair Prompt، Claim، Evidence و Semantic Failure Context هر Candidate بدون تغییر باقی می‌مانند.
</p>
<p>
Recovery در Namespace مستقل، با Fingerprint و Checkpoint مستقل انجام می‌شود و اجرای مجدد فقط Candidateهای فاقد Recovery Checkpoint معتبر را به API ارسال خواهد کرد. معیار موفقیت این مرحله دستیابی به ۳۷ Repair دارای Structural PASS و صفر API Failure است. پس از آن، فقط همین ۳۷ QA اصلاح‌شده وارد Semantic Revalidation مستقل خواهند شد.

````python
# Code Cell 52 — Structural Recovery of 7 Truncated Targeted QA Repairs

from collections import Counter
from concurrent.futures import ThreadPoolExecutor, as_completed
from pathlib import Path
import hashlib
import json
import os
import re
import tempfile
import threading
import time


# =========================================================
# CONFIG
# =========================================================

CELL52_RUNNER_VERSION = (
    "golden_v2_targeted_qa_semantic_repair_recovery_v1"
)

CELL52_GENERATION_PROTOCOL_VERSION = (
    CELL51_GENERATION_PROTOCOL_VERSION
)

CELL52_PROMPT_VERSION = (
    CELL51_PROMPT_VERSION
)

CELL52_OUTPUT_SCHEMA_VERSION = (
    CELL51_OUTPUT_SCHEMA_VERSION
)

CELL52_MAX_TOKENS = 3072
CELL52_TEMPERATURE = CELL51_TEMPERATURE
CELL52_TOP_P = CELL51_TOP_P

CELL52_MAX_ATTEMPTS = 2
CELL52_RETRY_SECONDS = 2.0

CELL52_MAX_WORKERS = 7
CELL52_REQUEST_START_SPACING_SECONDS = 0.6
CELL52_REQUEST_TIMEOUT_SECONDS = 180.0

# Before final Restart -> Run All this MUST become False.
CELL52_ALLOW_API_FOR_MISSING = False


# =========================================================
# REQUIRED STATE
# =========================================================

cell52_required_names = [
    "project_root",
    "semantic_review_client",
    "semantic_review_model",
    "GOLDEN_V2_GENERATION_SYSTEM_PROMPT_V4",
    "cell41_validate_generation_output_v4",
    "cell51_requests",
    "cell51_repair_results",
    "cell51_build_repair_packet",
    "cell51_build_user_prompt",
]

cell52_missing_state = [
    name
    for name in cell52_required_names
    if name not in globals()
]

assert not cell52_missing_state, (
    "Missing required Cell 52 state: "
    + ", ".join(cell52_missing_state)
)


# =========================================================
# EXACT CELL 51 STATE
# =========================================================

assert len(cell51_repair_results) == 37

cell52_cell51_by_candidate_id = {
    result["candidate_id"]: result
    for result in cell51_repair_results
}

cell52_cell51_request_by_candidate_id = {
    request["candidate_id"]: request
    for request in cell51_requests
}

assert len(cell52_cell51_by_candidate_id) == 37
assert len(cell52_cell51_request_by_candidate_id) == 37


cell52_preserved_repair_results = [
    result
    for result in cell51_repair_results
    if result.get("structural_status") == "PASS"
]

cell52_failed_repair_results = [
    result
    for result in cell51_repair_results
    if result.get("structural_status") != "PASS"
]


assert len(cell52_preserved_repair_results) == 30, (
    "Expected exactly 30 structurally valid Cell 51 repairs."
)

assert len(cell52_failed_repair_results) == 7, (
    "Expected exactly 7 Cell 51 structural failures."
)


cell52_recovery_candidate_ids = [
    result["candidate_id"]
    for result in cell52_failed_repair_results
]

assert len(set(cell52_recovery_candidate_ids)) == 7


for result in cell52_failed_repair_results:

    assert result.get("finish_reason") == "length", (
        "Cell 52 expected every recovery candidate "
        "to be a length-truncated Cell 51 result."
    )

    assert result.get("execution_status") == "COMPLETED", (
        "Cell 52 expected no Cell 51 API failures."
    )


# =========================================================
# HASH UTILITIES
# =========================================================

def cell52_canonical_json(value):

    return json.dumps(
        value,
        ensure_ascii=False,
        sort_keys=True,
        separators=(",", ":"),
    )


def cell52_sha256_text(text):

    return hashlib.sha256(
        text.encode("utf-8")
    ).hexdigest()


# =========================================================
# RECOVERY REQUEST CONTRACT
# =========================================================

def cell52_build_request(candidate_id):

    repair_packet = (
        cell51_build_repair_packet(
            candidate_id
        )
    )

    user_prompt = (
        cell51_build_user_prompt(
            candidate_id
        )
    )

    source_cell51_result = (
        cell52_cell51_by_candidate_id[
            candidate_id
        ]
    )

    source_failure = {
        "execution_status": (
            source_cell51_result.get(
                "execution_status"
            )
        ),
        "structural_status": (
            source_cell51_result.get(
                "structural_status"
            )
        ),
        "finish_reason": (
            source_cell51_result.get(
                "finish_reason"
            )
        ),
        "structural_errors": (
            source_cell51_result.get(
                "structural_errors",
                [],
            )
        ),
    }

    contract = {
        "runner_version": (
            CELL52_RUNNER_VERSION
        ),

        "generation_protocol_version": (
            CELL52_GENERATION_PROTOCOL_VERSION
        ),

        "prompt_version": (
            CELL52_PROMPT_VERSION
        ),

        "output_schema_version": (
            CELL52_OUTPUT_SCHEMA_VERSION
        ),

        "candidate_id": candidate_id,

        "repair_packet_sha256": (
            cell52_sha256_text(
                cell52_canonical_json(
                    repair_packet
                )
            )
        ),

        "source_cell51_failure": (
            source_failure
        ),

        "model": (
            semantic_review_model
        ),

        "system_prompt": (
            GOLDEN_V2_GENERATION_SYSTEM_PROMPT_V4
        ),

        "user_prompt": (
            user_prompt
        ),

        "temperature": (
            CELL52_TEMPERATURE
        ),

        "top_p": (
            CELL52_TOP_P
        ),

        "max_tokens": (
            CELL52_MAX_TOKENS
        ),

        "enable_thinking": False,
    }

    contract["request_fingerprint"] = (
        cell52_sha256_text(
            cell52_canonical_json(
                contract
            )
        )
    )

    return contract


cell52_requests = [
    cell52_build_request(candidate_id)
    for candidate_id
    in cell52_recovery_candidate_ids
]

assert len(cell52_requests) == 7

assert len({
    request["request_fingerprint"]
    for request in cell52_requests
}) == 7


# =========================================================
# CHECKPOINT NAMESPACE
# =========================================================

cell52_checkpoint_dir = (
    Path(project_root)
    / "artifacts"
    / "phase2"
    / "golden_dataset_v2"
    / "work"
    / "targeted_qa_semantic_repair_recovery_v1"
)

cell52_checkpoint_dir.mkdir(
    parents=True,
    exist_ok=True,
)


def cell52_checkpoint_path(candidate_id):

    safe_id = re.sub(
        r"[^A-Za-z0-9_.-]+",
        "_",
        candidate_id,
    )

    return (
        cell52_checkpoint_dir
        / f"{safe_id}.json"
    )


# =========================================================
# ATOMIC WRITE
# =========================================================

def cell52_atomic_write_json(path, payload):

    path = Path(path)

    content = (
        json.dumps(
            payload,
            ensure_ascii=False,
            sort_keys=True,
            indent=2,
        )
        + "\n"
    )

    fd, temp_name = tempfile.mkstemp(
        prefix=path.name + ".",
        suffix=".tmp",
        dir=str(path.parent),
    )

    try:

        with os.fdopen(
            fd,
            "w",
            encoding="utf-8",
        ) as handle:

            handle.write(content)
            handle.flush()
            os.fsync(handle.fileno())

        os.replace(
            temp_name,
            path,
        )

    except Exception:

        try:
            os.unlink(temp_name)
        except FileNotFoundError:
            pass

        raise


# =========================================================
# CHECKPOINT CONTRACT
# =========================================================

def cell52_build_checkpoint(request, result):

    checkpoint = {
        "checkpoint_version": (
            CELL52_RUNNER_VERSION
        ),

        "candidate_id": (
            request["candidate_id"]
        ),

        "request_fingerprint": (
            request["request_fingerprint"]
        ),

        "repair_packet_sha256": (
            request["repair_packet_sha256"]
        ),

        "generation_protocol_version": (
            CELL52_GENERATION_PROTOCOL_VERSION
        ),

        "prompt_version": (
            CELL52_PROMPT_VERSION
        ),

        "output_schema_version": (
            CELL52_OUTPUT_SCHEMA_VERSION
        ),

        "model": (
            semantic_review_model
        ),

        "temperature": (
            CELL52_TEMPERATURE
        ),

        "top_p": (
            CELL52_TOP_P
        ),

        "max_tokens": (
            CELL52_MAX_TOKENS
        ),

        "result": result,
    }

    checkpoint[
        "checkpoint_payload_sha256"
    ] = cell52_sha256_text(
        cell52_canonical_json(
            checkpoint
        )
    )

    return checkpoint


def cell52_validate_checkpoint(
    checkpoint,
    request,
):

    errors = []

    if not isinstance(checkpoint, dict):
        return ["CHECKPOINT_NOT_OBJECT"]

    stored_sha = checkpoint.get(
        "checkpoint_payload_sha256"
    )

    payload_without_sha = {
        key: value
        for key, value in checkpoint.items()
        if key != "checkpoint_payload_sha256"
    }

    recalculated_sha = (
        cell52_sha256_text(
            cell52_canonical_json(
                payload_without_sha
            )
        )
    )

    if stored_sha != recalculated_sha:
        errors.append(
            "CHECKPOINT_HASH_MISMATCH"
        )

    expected_values = {
        "checkpoint_version": (
            CELL52_RUNNER_VERSION
        ),
        "candidate_id": (
            request["candidate_id"]
        ),
        "request_fingerprint": (
            request["request_fingerprint"]
        ),
        "repair_packet_sha256": (
            request["repair_packet_sha256"]
        ),
        "generation_protocol_version": (
            CELL52_GENERATION_PROTOCOL_VERSION
        ),
        "prompt_version": (
            CELL52_PROMPT_VERSION
        ),
        "output_schema_version": (
            CELL52_OUTPUT_SCHEMA_VERSION
        ),
        "model": semantic_review_model,
        "temperature": (
            CELL52_TEMPERATURE
        ),
        "top_p": (
            CELL52_TOP_P
        ),
        "max_tokens": (
            CELL52_MAX_TOKENS
        ),
    }

    for field_name, expected_value in (
        expected_values.items()
    ):

        if checkpoint.get(
            field_name
        ) != expected_value:

            errors.append(
                f"{field_name.upper()}_MISMATCH"
            )

    result = checkpoint.get("result")

    if not isinstance(result, dict):

        errors.append(
            "CHECKPOINT_RESULT_INVALID"
        )

    elif result.get(
        "candidate_id"
    ) != request["candidate_id"]:

        errors.append(
            "CHECKPOINT_RESULT_CANDIDATE_MISMATCH"
        )

    return sorted(set(errors))


def cell52_load_checkpoint(request):

    path = cell52_checkpoint_path(
        request["candidate_id"]
    )

    if not path.exists():

        return None, "MISSING", []

    try:

        with path.open(
            "r",
            encoding="utf-8",
        ) as handle:

            checkpoint = json.load(handle)

    except Exception as exc:

        return (
            None,
            "UNREADABLE",
            [
                f"{type(exc).__name__}: {exc}"
            ],
        )

    errors = (
        cell52_validate_checkpoint(
            checkpoint,
            request,
        )
    )

    if errors:

        return (
            checkpoint,
            "INVALID",
            errors,
        )

    return (
        checkpoint,
        "VALID",
        [],
    )


# =========================================================
# RESPONSE UTILITIES
# =========================================================

def cell52_clean_model_json(content):

    if not isinstance(content, str):
        return ""

    clean = content.strip()

    if clean.startswith("```"):

        clean = re.sub(
            r"^```(?:json)?\s*",
            "",
            clean,
            flags=re.IGNORECASE,
        )

        clean = re.sub(
            r"\s*```$",
            "",
            clean,
        )

    return clean.strip()


def cell52_is_rate_limit_exception(exc):

    error_text = str(exc).lower()
    error_type = type(exc).__name__.lower()

    return (
        "ratelimit" in error_type
        or "rate_limit" in error_type
        or "429" in error_text
        or "too many requests" in error_text
    )


# =========================================================
# REQUEST THROTTLE
# =========================================================

cell52_request_start_lock = (
    threading.Lock()
)

cell52_last_request_start = [0.0]


def cell52_wait_for_request_slot():

    with cell52_request_start_lock:

        now = time.monotonic()

        elapsed = (
            now
            - cell52_last_request_start[0]
        )

        wait_seconds = max(
            0.0,
            CELL52_REQUEST_START_SPACING_SECONDS
            - elapsed,
        )

        if wait_seconds > 0:
            time.sleep(wait_seconds)

        cell52_last_request_start[0] = (
            time.monotonic()
        )


# =========================================================
# COST
# =========================================================

cell52_has_cost_rates = (
    "INPUT_TOMAN_PER_TOKEN" in globals()
    and
    "OUTPUT_TOMAN_PER_TOKEN" in globals()
)


def cell52_estimated_cost_toman(
    input_tokens,
    output_tokens,
):

    if not cell52_has_cost_rates:
        return None

    return (
        input_tokens
        * INPUT_TOMAN_PER_TOKEN
        +
        output_tokens
        * OUTPUT_TOMAN_PER_TOKEN
    )


# =========================================================
# API FAILURE RESULT
# =========================================================

def cell52_build_api_failure_result(
    request,
    api_attempts,
    attempt_history,
):

    return {
        "candidate_id": (
            request["candidate_id"]
        ),

        "request_fingerprint": (
            request["request_fingerprint"]
        ),

        "execution_status": (
            "API_FAILED"
        ),

        "api_attempts": api_attempts,

        "attempt_history": (
            attempt_history
        ),

        "finish_reason": None,

        "usage": {
            "input_tokens": 0,
            "output_tokens": 0,
            "reasoning_tokens": 0,
            "total_tokens": 0,
        },

        "estimated_cost_toman": None,

        "raw_content": None,
        "parsed_json": None,
        "parse_error": None,

        "structural_errors": [
            "API_FAILED"
        ],

        "structural_status": "FAIL",
    }


# =========================================================
# EXECUTION
# =========================================================

def cell52_execute_request(request):

    attempt_history = []
    api_attempts = 0

    for attempt_number in range(
        1,
        CELL52_MAX_ATTEMPTS + 1,
    ):

        cell52_wait_for_request_slot()

        api_attempts += 1

        try:

            response = (
                semantic_review_client
                .chat
                .completions
                .create(
                    model=semantic_review_model,

                    messages=[
                        {
                            "role": "system",
                            "content": (
                                GOLDEN_V2_GENERATION_SYSTEM_PROMPT_V4
                            ),
                        },
                        {
                            "role": "user",
                            "content": (
                                request["user_prompt"]
                            ),
                        },
                    ],

                    temperature=(
                        CELL52_TEMPERATURE
                    ),

                    top_p=(
                        CELL52_TOP_P
                    ),

                    max_tokens=(
                        CELL52_MAX_TOKENS
                    ),

                    timeout=(
                        CELL52_REQUEST_TIMEOUT_SECONDS
                    ),

                    extra_body={
                        "enable_thinking": False,
                    },
                )
            )

            if response is None:
                raise RuntimeError(
                    "EMPTY_API_RESPONSE"
                )

            choices = getattr(
                response,
                "choices",
                None,
            )

            if (
                not isinstance(
                    choices,
                    (list, tuple),
                )
                or len(choices) == 0
            ):

                raise RuntimeError(
                    "API_RESPONSE_CHOICES_MISSING_OR_EMPTY"
                )

            choice = choices[0]

            if choice is None:
                raise RuntimeError(
                    "API_RESPONSE_FIRST_CHOICE_IS_NONE"
                )

            message = getattr(
                choice,
                "message",
                None,
            )

            if message is None:
                raise RuntimeError(
                    "API_RESPONSE_MESSAGE_MISSING"
                )

        except Exception as exc:

            attempt_history.append({
                "attempt": attempt_number,
                "status": "API_ERROR",
                "error": (
                    f"{type(exc).__name__}: {exc}"
                ),
            })

            if (
                attempt_number
                < CELL52_MAX_ATTEMPTS
            ):

                wait_seconds = (
                    70.0
                    if cell52_is_rate_limit_exception(
                        exc
                    )
                    else (
                        CELL52_RETRY_SECONDS
                        * attempt_number
                    )
                )

                time.sleep(wait_seconds)
                continue

            return (
                cell52_build_api_failure_result(
                    request,
                    api_attempts,
                    attempt_history,
                )
            )

        finish_reason = getattr(
            choice,
            "finish_reason",
            None,
        )

        raw_content = getattr(
            message,
            "content",
            None,
        )

        usage = getattr(
            response,
            "usage",
            None,
        )

        input_tokens = int(
            getattr(
                usage,
                "prompt_tokens",
                0,
            )
            or 0
        )

        output_tokens = int(
            getattr(
                usage,
                "completion_tokens",
                0,
            )
            or 0
        )

        total_tokens = int(
            getattr(
                usage,
                "total_tokens",
                0,
            )
            or (
                input_tokens
                + output_tokens
            )
        )

        reasoning_tokens = 0

        completion_details = getattr(
            usage,
            "completion_tokens_details",
            None,
        )

        if completion_details is not None:

            reasoning_tokens = int(
                getattr(
                    completion_details,
                    "reasoning_tokens",
                    0,
                )
                or 0
            )

        clean_content = (
            cell52_clean_model_json(
                raw_content
            )
        )

        parsed_json = None
        parse_error = None

        try:

            parsed_json = json.loads(
                clean_content
            )

        except Exception as exc:

            parse_error = (
                f"{type(exc).__name__}: {exc}"
            )

        if parsed_json is None:

            structural_errors = [
                "JSON_PARSE_FAILED"
            ]

        else:

            structural_errors = (
                cell41_validate_generation_output_v4(
                    parsed_json
                )
            )

        structural_status = (
            "PASS"
            if not structural_errors
            else "FAIL"
        )

        return {
            "candidate_id": (
                request["candidate_id"]
            ),

            "request_fingerprint": (
                request["request_fingerprint"]
            ),

            "execution_status": (
                "COMPLETED"
            ),

            "api_attempts": (
                api_attempts
            ),

            "attempt_history": (
                attempt_history
            ),

            "finish_reason": (
                finish_reason
            ),

            "usage": {
                "input_tokens": (
                    input_tokens
                ),
                "output_tokens": (
                    output_tokens
                ),
                "reasoning_tokens": (
                    reasoning_tokens
                ),
                "total_tokens": (
                    total_tokens
                ),
            },

            "estimated_cost_toman": (
                cell52_estimated_cost_toman(
                    input_tokens,
                    output_tokens,
                )
            ),

            "raw_content": raw_content,
            "parsed_json": parsed_json,
            "parse_error": parse_error,

            "structural_errors": (
                structural_errors
            ),

            "structural_status": (
                structural_status
            ),
        }

    raise AssertionError(
        "Unexpected Cell 52 execution path."
    )


# =========================================================
# PREFLIGHT CHECKPOINT SCAN
# =========================================================

cell52_valid_before_api = {}
cell52_missing_requests = []
cell52_invalid_before_api = []


for request in cell52_requests:

    (
        checkpoint,
        status,
        errors,
    ) = cell52_load_checkpoint(
        request
    )

    candidate_id = (
        request["candidate_id"]
    )

    if status == "VALID":

        cell52_valid_before_api[
            candidate_id
        ] = checkpoint

    elif status == "MISSING":

        cell52_missing_requests.append(
            request
        )

    else:

        cell52_invalid_before_api.append({
            "candidate_id": candidate_id,
            "status": status,
            "errors": errors,
        })


assert not cell52_invalid_before_api, (
    "Invalid/unreadable Cell 52 checkpoints detected. "
    "Refusing automatic paid replacement. "
    f"Examples: {cell52_invalid_before_api[:5]}"
)


# =========================================================
# PREFLIGHT OUTPUT
# =========================================================

print(
    "=== GOLDEN V2 TARGETED QA REPAIR "
    "STRUCTURAL RECOVERY PREFLIGHT ==="
)

print()

print(
    "Original repair population:",
    37,
)

print(
    "Preserved Cell 51 structural PASS:",
    len(
        cell52_preserved_repair_results
    ),
)

print(
    "Recovery population:",
    len(cell52_requests),
)

print(
    "Exact valid recovery checkpoints:",
    len(cell52_valid_before_api),
)

print(
    "Missing recovery checkpoints:",
    len(cell52_missing_requests),
)

print(
    "Model:",
    semantic_review_model,
)

print(
    "Generation protocol:",
    CELL52_GENERATION_PROTOCOL_VERSION,
)

print(
    "Repair prompt version:",
    CELL52_PROMPT_VERSION,
)

print(
    "Recovery max tokens:",
    CELL52_MAX_TOKENS,
)

print(
    "Parallel workers:",
    CELL52_MAX_WORKERS,
)

print(
    "API allowed for missing exact checkpoints:",
    CELL52_ALLOW_API_FOR_MISSING,
)

print(
    "Previously valid Cell 51 repairs sent to API:",
    0,
)

print(
    "Previously accepted 3776 QA sent to API:",
    0,
)

print()


if (
    cell52_missing_requests
    and
    not CELL52_ALLOW_API_FOR_MISSING
):

    raise AssertionError(
        "Cell 52 has missing recovery checkpoints "
        "but API execution is disabled."
    )


# =========================================================
# EXECUTE + CHECKPOINT
# =========================================================

cell52_api_candidates_this_execution = 0
cell52_api_attempts_this_execution = 0
cell52_checkpoints_created_this_execution = 0


def cell52_execute_and_checkpoint(request):

    candidate_id = (
        request["candidate_id"]
    )

    result = (
        cell52_execute_request(
            request
        )
    )

    checkpoint = (
        cell52_build_checkpoint(
            request,
            result,
        )
    )

    cell52_atomic_write_json(
        cell52_checkpoint_path(
            candidate_id
        ),
        checkpoint,
    )

    (
        verified_checkpoint,
        verified_status,
        verified_errors,
    ) = cell52_load_checkpoint(
        request
    )

    if verified_status != "VALID":

        raise AssertionError(
            "Cell 52 checkpoint verification failed "
            f"for {candidate_id}: "
            f"{verified_errors}"
        )

    return (
        candidate_id,
        verified_checkpoint["result"],
    )


if cell52_missing_requests:

    completed_count = 0

    executor = ThreadPoolExecutor(
        max_workers=CELL52_MAX_WORKERS
    )

    future_to_request = {}

    try:

        future_to_request = {
            executor.submit(
                cell52_execute_and_checkpoint,
                request,
            ): request

            for request
            in cell52_missing_requests
        }

        for future in as_completed(
            future_to_request
        ):

            request = (
                future_to_request[future]
            )

            candidate_id = (
                request["candidate_id"]
            )

            try:

                (
                    resolved_candidate_id,
                    result,
                ) = future.result()

            except Exception as exc:

                completed_count += 1

                print(
                    f"[{completed_count:02d}/"
                    f"{len(cell52_missing_requests):02d}]",
                    candidate_id,
                    "| WORKER_EXCEPTION =",
                    f"{type(exc).__name__}: {exc}",
                    flush=True,
                )

                continue

            assert (
                resolved_candidate_id
                == candidate_id
            )

            completed_count += 1

            cell52_api_candidates_this_execution += 1

            cell52_api_attempts_this_execution += int(
                result.get(
                    "api_attempts",
                    0,
                )
                or 0
            )

            cell52_checkpoints_created_this_execution += 1

            print(
                f"[{completed_count:02d}/"
                f"{len(cell52_missing_requests):02d}]",
                candidate_id,
                "| structural=",
                result.get(
                    "structural_status"
                ),
                "| finish=",
                result.get(
                    "finish_reason"
                ),
                flush=True,
            )

    except KeyboardInterrupt:

        print(
            "\nCell 52 interrupted by user. "
            "Completed recovery checkpoints remain preserved.",
            flush=True,
        )

        for future in future_to_request:

            if not future.done():
                future.cancel()

        raise

    finally:

        executor.shutdown(
            wait=False,
            cancel_futures=True,
        )


# =========================================================
# FINAL RECOVERY CHECKPOINT RESCAN
# =========================================================

cell52_recovery_results = []

cell52_valid_checkpoint_count = 0
cell52_missing_checkpoint_count = 0
cell52_invalid_checkpoint_count = 0


for request in cell52_requests:

    (
        checkpoint,
        status,
        errors,
    ) = cell52_load_checkpoint(
        request
    )

    if status == "VALID":

        cell52_valid_checkpoint_count += 1

        cell52_recovery_results.append(
            checkpoint["result"]
        )

    elif status == "MISSING":

        cell52_missing_checkpoint_count += 1

    else:

        cell52_invalid_checkpoint_count += 1


# =========================================================
# MERGE 30 PRESERVED + 7 RECOVERED
# =========================================================

cell52_recovery_by_candidate_id = {
    result["candidate_id"]: result
    for result in cell52_recovery_results
}

cell52_final_repair_results = []


for request in cell51_requests:

    candidate_id = (
        request["candidate_id"]
    )

    if candidate_id in (
        cell52_recovery_by_candidate_id
    ):

        cell52_final_repair_results.append(
            cell52_recovery_by_candidate_id[
                candidate_id
            ]
        )

    else:

        cell52_final_repair_results.append(
            cell52_cell51_by_candidate_id[
                candidate_id
            ]
        )


assert len(
    cell52_final_repair_results
) == 37


# =========================================================
# DISTRIBUTIONS
# =========================================================

cell52_recovery_structural_distribution = Counter(
    result.get("structural_status")
    for result in cell52_recovery_results
)

cell52_recovery_finish_distribution = Counter(
    result.get("finish_reason")
    for result in cell52_recovery_results
)

cell52_final_structural_distribution = Counter(
    result.get("structural_status")
    for result in cell52_final_repair_results
)

cell52_final_execution_distribution = Counter(
    result.get("execution_status")
    for result in cell52_final_repair_results
)

cell52_final_finish_distribution = Counter(
    result.get("finish_reason")
    for result in cell52_final_repair_results
)


cell52_final_structural_pass_count = (
    cell52_final_structural_distribution.get(
        "PASS",
        0,
    )
)

cell52_final_structural_fail_count = (
    cell52_final_structural_distribution.get(
        "FAIL",
        0,
    )
)

cell52_final_api_failure_count = (
    cell52_final_execution_distribution.get(
        "API_FAILED",
        0,
    )
)


# =========================================================
# COST
# =========================================================

cell52_known_costs = [
    result.get(
        "estimated_cost_toman"
    )
    for result in cell52_recovery_results
    if isinstance(
        result.get(
            "estimated_cost_toman"
        ),
        (int, float),
    )
]

cell52_estimated_recovery_cost_toman = (
    sum(cell52_known_costs)
    if cell52_known_costs
    else None
)


# =========================================================
# RUNNER HEALTH
# =========================================================

cell52_checks = {
    "cell51_repair_population_37": (
        len(cell51_repair_results)
        == 37
    ),

    "preserved_cell51_pass_30": (
        len(
            cell52_preserved_repair_results
        )
        == 30
    ),

    "recovery_population_7": (
        len(cell52_requests)
        == 7
    ),

    "all_source_failures_were_length": (
        all(
            result.get("finish_reason")
            == "length"
            for result
            in cell52_failed_repair_results
        )
    ),

    "all_recovery_checkpoints_present": (
        cell52_valid_checkpoint_count
        == 7
    ),

    "zero_missing_recovery_checkpoints": (
        cell52_missing_checkpoint_count
        == 0
    ),

    "zero_invalid_recovery_checkpoints": (
        cell52_invalid_checkpoint_count
        == 0
    ),

    "final_repair_population_37": (
        len(
            cell52_final_repair_results
        )
        == 37
    ),

    "final_candidate_identity_exact": (
        set(
            result["candidate_id"]
            for result
            in cell52_final_repair_results
        )
        ==
        set(
            cell50_repair_candidate_ids
        )
    ),

    "previously_valid_30_not_regenerated": True,

    "previously_accepted_3776_not_regenerated": True,
}


cell52_runner_status = (
    "PASS"
    if all(cell52_checks.values())
    else "FAIL"
)


if cell52_runner_status != "PASS":

    cell52_next_readiness = (
        "BLOCKED_CELL52_RUNNER_REPAIR_REQUIRED"
    )

elif cell52_final_api_failure_count > 0:

    cell52_next_readiness = (
        "READY_FOR_REMAINING_QA_REPAIR_API_RECOVERY"
    )

elif cell52_final_structural_fail_count > 0:

    cell52_next_readiness = (
        "READY_FOR_REMAINING_QA_REPAIR_STRUCTURAL_RECOVERY"
    )

elif cell52_final_structural_pass_count == 37:

    cell52_next_readiness = (
        "READY_FOR_TARGETED_QA_REPAIR_SEMANTIC_REVALIDATION"
    )

else:

    cell52_next_readiness = (
        "READY_FOR_CELL52_RESULT_ANALYSIS"
    )


# =========================================================
# OUTPUT
# =========================================================

print()

print(
    "=== GOLDEN V2 TARGETED QA REPAIR "
    "STRUCTURAL RECOVERY RESULT ==="
)

print()

print(
    "Recovery population:",
    len(cell52_requests),
)

print(
    "Valid recovery checkpoints:",
    cell52_valid_checkpoint_count,
)

print(
    "Missing recovery checkpoints:",
    cell52_missing_checkpoint_count,
)

print(
    "Invalid recovery checkpoints:",
    cell52_invalid_checkpoint_count,
)

print()

print(
    "API candidates this execution:",
    cell52_api_candidates_this_execution,
)

print(
    "API attempts this execution:",
    cell52_api_attempts_this_execution,
)

print(
    "Recovery checkpoints created this execution:",
    cell52_checkpoints_created_this_execution,
)

print()

print(
    "Recovery structural distribution:",
    dict(
        cell52_recovery_structural_distribution
    ),
)

print(
    "Recovery finish distribution:",
    dict(
        cell52_recovery_finish_distribution
    ),
)

print(
    "Estimated recovery cost (toman):",
    cell52_estimated_recovery_cost_toman,
)

print()

print(
    "=== FINAL 37 REPAIR STATE ==="
)

print(
    "Repair population:",
    len(
        cell52_final_repair_results
    ),
)

print(
    "Structural distribution:",
    dict(
        cell52_final_structural_distribution
    ),
)

print(
    "Structural PASS:",
    cell52_final_structural_pass_count,
    "/ 37",
)

print(
    "Structural FAIL:",
    cell52_final_structural_fail_count,
    "/ 37",
)

print(
    "API failures:",
    cell52_final_api_failure_count,
)

print(
    "Finish reason distribution:",
    dict(
        cell52_final_finish_distribution
    ),
)

print()

print("Runner checks:")

for check_name, check_value in (
    cell52_checks.items()
):

    print(
        " -",
        check_name,
        ":",
        (
            "PASS"
            if check_value
            else "FAIL"
        ),
    )

print()

print(
    "Runner health:",
    cell52_runner_status,
)

print(
    "Next readiness:",
    cell52_next_readiness,
)

print(
    "CELL 52 CHECKPOINTED RUNNER:",
    cell52_runner_status,
)


assert cell52_runner_status == "PASS", (
    "Cell 52 checkpointed runner health failed."
)
````

<p>
پس از تکمیل Recovery ساختاری، هر ۳۷ Question/Answer اصلاح‌شده اکنون دارای خروجی ساختاری معتبر هستند و هیچ API Failure یا پاسخ ناقصی در جمعیت Repair باقی نمانده است. ۳۷۷۶ Question/Answer دارای Semantic PASS قبلی نیز بدون تغییر محفوظ مانده‌اند.
</p>
<p>
در این مرحله فقط همین ۳۷ QA اصلاح‌شده با همان Semantic Validation Protocol مرحله ۴۸ در برابر Claim و Evidence متناظر ارزیابی می‌شوند. Validator مستقل از Generator عمل می‌کند و هشت بُعد شامل هدف‌گیری صحیح Claim، بازیابی معنای Claim در Answer، وفاداری Scope، حفظ Modality و Quantifier، مستقل‌بودن Question، نبود Neighbor Fact Drift، نبود External Knowledge و هم‌راستایی Answer با Question را بررسی خواهد کرد.

این اجرا Checkpoint-first، Fingerprint-bound و قابل Resume است و ۳۷۷۶ مورد Semantic PASS قبلی هرگز دوباره به API ارسال نمی‌شوند. در پایان، Judgmentهای جدید این ۳۷ مورد با ۳۷۷۶ Judgment سالم قبلی ادغام می‌شوند تا وضعیت معنایی نهایی تمام ۳۸۱۳ Candidate مشخص شود. مواردی که همچنان Semantic FAIL باشند در مرحله بعد فقط به‌صورت هدفمند تعیین تکلیف خواهند شد.
</p>

````python
# Code Cell 53 — Independent Semantic Revalidation of 37 Repaired QAs

from collections import Counter
from concurrent.futures import ThreadPoolExecutor, as_completed
from pathlib import Path
import hashlib
import json
import os
import re
import tempfile
import threading
import time


# =========================================================
# CONFIG
# =========================================================

CELL53_RUNNER_VERSION = (
    "golden_v2_targeted_qa_repair_semantic_revalidation_v1"
)

CELL53_VALIDATION_PROTOCOL_VERSION = (
    CELL49_VALIDATION_PROTOCOL_VERSION
)

CELL53_OUTPUT_SCHEMA_VERSION = (
    CELL49_OUTPUT_SCHEMA_VERSION
)

CELL53_MAX_TOKENS = 3072
CELL53_TEMPERATURE = 0.0
CELL53_TOP_P = 1.0

CELL53_MAX_ATTEMPTS = 2
CELL53_RETRY_SECONDS = 2.0

CELL53_MAX_WORKERS = 16
CELL53_REQUEST_START_SPACING_SECONDS = 0.6
CELL53_REQUEST_TIMEOUT_SECONDS = 180.0

# Before final Restart -> Run All this MUST become False.
CELL53_ALLOW_API_FOR_MISSING = False


# =========================================================
# REQUIRED STATE
# =========================================================

cell53_required_names = [
    "project_root",
    "semantic_review_client",
    "semantic_review_model",
    "cell25_candidate_skeletons",
    "cell49_final_semantic_results",
    "cell50_repair_candidate_ids",
    "cell52_final_repair_results",
    "CELL37_SYSTEM_PROMPT",
    "CELL37_BOOLEAN_FIELDS",
    "cell37_validate_output",
]

cell53_missing_state = [
    name
    for name in cell53_required_names
    if name not in globals()
]

assert not cell53_missing_state, (
    "Missing required Cell 53 state: "
    + ", ".join(cell53_missing_state)
)


# =========================================================
# EXACT INPUT STATE
# =========================================================

assert len(
    cell49_final_semantic_results
) == 3813

assert len(
    cell50_repair_candidate_ids
) == 37

assert len(
    cell52_final_repair_results
) == 37


cell53_skeleton_by_candidate_id = {
    record["candidate_id"]: record
    for record in cell25_candidate_skeletons
}

cell53_previous_semantic_by_candidate_id = {
    result["candidate_id"]: result
    for result in cell49_final_semantic_results
}

cell53_repair_by_candidate_id = {
    result["candidate_id"]: result
    for result in cell52_final_repair_results
}


assert (
    set(
        cell53_repair_by_candidate_id
    )
    ==
    set(
        cell50_repair_candidate_ids
    )
)

assert all(
    result.get("structural_status") == "PASS"
    for result
    in cell52_final_repair_results
)


# =========================================================
# PRESERVED 3776
# =========================================================

cell53_preserved_semantic_results = [
    result
    for result
    in cell49_final_semantic_results
    if result["candidate_id"]
    not in set(
        cell50_repair_candidate_ids
    )
]

assert len(
    cell53_preserved_semantic_results
) == 3776

assert all(
    result.get("structural_status") == "PASS"
    and
    result.get("semantic_verdict") == "PASS"
    for result
    in cell53_preserved_semantic_results
)


# =========================================================
# HASH UTILITIES
# =========================================================

def cell53_canonical_json(value):

    return json.dumps(
        value,
        ensure_ascii=False,
        sort_keys=True,
        separators=(",", ":"),
    )


def cell53_sha256_text(text):

    return hashlib.sha256(
        text.encode("utf-8")
    ).hexdigest()


# =========================================================
# VALIDATION PACKET
# =========================================================

def cell53_build_validation_packet(
    candidate_id,
):

    skeleton = (
        cell53_skeleton_by_candidate_id[
            candidate_id
        ]
    )

    repaired_result = (
        cell53_repair_by_candidate_id[
            candidate_id
        ]
    )

    repaired_qa = (
        repaired_result.get(
            "parsed_json"
        )
    )

    assert isinstance(
        repaired_qa,
        dict,
    )

    evidence_ids = list(
        skeleton["evidence_ids"]
    )

    evidence_quotes = list(
        skeleton["evidence_quotes"]
    )

    assert len(
        evidence_ids
    ) == len(
        evidence_quotes
    )

    evidence = []

    for index, (
        evidence_id,
        evidence_quote,
    ) in enumerate(
        zip(
            evidence_ids,
            evidence_quotes,
        ),
        start=1,
    ):

        evidence.append({
            "index": index,
            "evidence_id": evidence_id,
            "quote": evidence_quote,
        })

    return {
        "candidate_id": candidate_id,

        "claim": {
            "claim_id": (
                skeleton["claim_id"]
            ),

            "claim_text_fa": (
                skeleton["claim_text_fa"]
            ),

            "claim_type": (
                skeleton.get(
                    "claim_type"
                )
            ),

            "claim_types": (
                skeleton.get(
                    "claim_types",
                    [],
                )
            ),

            "authority_class": (
                skeleton.get(
                    "authority_class"
                )
            ),
        },

        "evidence": evidence,

        "generated_qa": {
            "question": (
                repaired_qa["question"]
            ),

            "expected_answer": (
                repaired_qa[
                    "expected_answer"
                ]
            ),

            "question_intent": (
                repaired_qa[
                    "question_intent"
                ]
            ),
        },
    }


# =========================================================
# SAME SEMANTIC VALIDATION TASK
# =========================================================

def cell53_build_user_prompt(
    candidate_id,
):

    packet = (
        cell53_build_validation_packet(
            candidate_id
        )
    )

    return (
        "Validate ONLY the generated Question and Expected Answer "
        "against the supplied prevalidated Claim and Evidence.\n\n"

        "The Claim and Evidence have already passed upstream semantic "
        "validation. Do NOT decide whether the Claim itself should be "
        "accepted or rejected. Do NOT perform a new Claim-Evidence "
        "eligibility review.\n\n"

        "Evaluate exactly these QA dimensions:\n"
        "1. question_targets_claim\n"
        "2. answer_recovers_claim\n"
        "3. scope_fidelity\n"
        "4. modality_quantifier_fidelity\n"
        "5. standalone_question\n"
        "6. no_neighbor_fact_drift\n"
        "7. no_external_knowledge\n"
        "8. answer_question_alignment\n\n"

        "Important fidelity rules:\n"
        "- The Question must ask for information actually asserted "
        "by the Claim.\n"
        "- The Expected Answer must correctly answer that Question "
        "and recover the relevant meaning of the Claim.\n"
        "- Preserve meaning-bearing negation, modality, attribution, "
        "conditions, exceptions, scope, quantities, percentages, "
        "comparisons and quantifiers such as only/تنها/فقط when "
        "they are relevant to the generated QA target.\n"
        "- Do not permit WHY/HOW/cause/reason/mechanism claims unless "
        "the supplied Claim actually asserts them.\n"
        "- Do not allow facts imported from neighboring Evidence text "
        "when those facts are not part of the target Claim.\n"
        "- Do not require external knowledge.\n"
        "- A narrower Question/Answer may be valid only when it "
        "faithfully targets a self-contained asserted part of the "
        "Claim without changing its meaning or introducing drift.\n\n"

        "Return only the JSON object required by the validation "
        "system prompt.\n\n"

        "VALIDATION PACKET:\n"
        + json.dumps(
            packet,
            ensure_ascii=False,
            sort_keys=True,
            indent=2,
        )
    )


# =========================================================
# REQUEST CONTRACT
# =========================================================

def cell53_build_request(
    candidate_id,
):

    packet = (
        cell53_build_validation_packet(
            candidate_id
        )
    )

    user_prompt = (
        cell53_build_user_prompt(
            candidate_id
        )
    )

    repaired_qa = (
        cell53_repair_by_candidate_id[
            candidate_id
        ]["parsed_json"]
    )

    contract = {
        "runner_version": (
            CELL53_RUNNER_VERSION
        ),

        "validation_protocol_version": (
            CELL53_VALIDATION_PROTOCOL_VERSION
        ),

        "output_schema_version": (
            CELL53_OUTPUT_SCHEMA_VERSION
        ),

        "candidate_id": candidate_id,

        "repaired_qa_sha256": (
            cell53_sha256_text(
                cell53_canonical_json(
                    repaired_qa
                )
            )
        ),

        "validation_packet_sha256": (
            cell53_sha256_text(
                cell53_canonical_json(
                    packet
                )
            )
        ),

        "model": semantic_review_model,

        "system_prompt": (
            CELL37_SYSTEM_PROMPT
        ),

        "user_prompt": (
            user_prompt
        ),

        "temperature": (
            CELL53_TEMPERATURE
        ),

        "top_p": (
            CELL53_TOP_P
        ),

        "max_tokens": (
            CELL53_MAX_TOKENS
        ),

        "enable_thinking": False,
    }

    contract["request_fingerprint"] = (
        cell53_sha256_text(
            cell53_canonical_json(
                contract
            )
        )
    )

    return contract


cell53_requests = [
    cell53_build_request(
        candidate_id
    )
    for candidate_id
    in cell50_repair_candidate_ids
]

assert len(cell53_requests) == 37

assert len({
    request["request_fingerprint"]
    for request
    in cell53_requests
}) == 37


# =========================================================
# CHECKPOINT NAMESPACE
# =========================================================

cell53_checkpoint_dir = (
    Path(project_root)
    / "artifacts"
    / "phase2"
    / "golden_dataset_v2"
    / "work"
    / "targeted_qa_repair_semantic_revalidation_v1"
)

cell53_checkpoint_dir.mkdir(
    parents=True,
    exist_ok=True,
)


def cell53_checkpoint_path(
    candidate_id,
):

    safe_id = re.sub(
        r"[^A-Za-z0-9_.-]+",
        "_",
        candidate_id,
    )

    return (
        cell53_checkpoint_dir
        / f"{safe_id}.json"
    )


# =========================================================
# ATOMIC WRITE
# =========================================================

def cell53_atomic_write_json(
    path,
    payload,
):

    path = Path(path)

    content = (
        json.dumps(
            payload,
            ensure_ascii=False,
            sort_keys=True,
            indent=2,
        )
        + "\n"
    )

    fd, temp_name = tempfile.mkstemp(
        prefix=path.name + ".",
        suffix=".tmp",
        dir=str(path.parent),
    )

    try:

        with os.fdopen(
            fd,
            "w",
            encoding="utf-8",
        ) as handle:

            handle.write(content)
            handle.flush()
            os.fsync(handle.fileno())

        os.replace(
            temp_name,
            path,
        )

    except Exception:

        try:
            os.unlink(temp_name)
        except FileNotFoundError:
            pass

        raise


# =========================================================
# CHECKPOINT CONTRACT
# =========================================================

def cell53_build_checkpoint(
    request,
    result,
):

    checkpoint = {
        "checkpoint_version": (
            CELL53_RUNNER_VERSION
        ),

        "candidate_id": (
            request["candidate_id"]
        ),

        "request_fingerprint": (
            request["request_fingerprint"]
        ),

        "repaired_qa_sha256": (
            request["repaired_qa_sha256"]
        ),

        "validation_packet_sha256": (
            request[
                "validation_packet_sha256"
            ]
        ),

        "validation_protocol_version": (
            CELL53_VALIDATION_PROTOCOL_VERSION
        ),

        "output_schema_version": (
            CELL53_OUTPUT_SCHEMA_VERSION
        ),

        "model": semantic_review_model,

        "temperature": (
            CELL53_TEMPERATURE
        ),

        "top_p": (
            CELL53_TOP_P
        ),

        "max_tokens": (
            CELL53_MAX_TOKENS
        ),

        "result": result,
    }

    checkpoint[
        "checkpoint_payload_sha256"
    ] = cell53_sha256_text(
        cell53_canonical_json(
            checkpoint
        )
    )

    return checkpoint


def cell53_validate_checkpoint(
    checkpoint,
    request,
):

    errors = []

    if not isinstance(
        checkpoint,
        dict,
    ):

        return [
            "CHECKPOINT_NOT_OBJECT"
        ]

    stored_sha = checkpoint.get(
        "checkpoint_payload_sha256"
    )

    payload_without_sha = {
        key: value
        for key, value
        in checkpoint.items()
        if key
        != "checkpoint_payload_sha256"
    }

    recalculated_sha = (
        cell53_sha256_text(
            cell53_canonical_json(
                payload_without_sha
            )
        )
    )

    if stored_sha != recalculated_sha:

        errors.append(
            "CHECKPOINT_HASH_MISMATCH"
        )

    expected_values = {
        "checkpoint_version": (
            CELL53_RUNNER_VERSION
        ),

        "candidate_id": (
            request["candidate_id"]
        ),

        "request_fingerprint": (
            request["request_fingerprint"]
        ),

        "repaired_qa_sha256": (
            request["repaired_qa_sha256"]
        ),

        "validation_packet_sha256": (
            request[
                "validation_packet_sha256"
            ]
        ),

        "validation_protocol_version": (
            CELL53_VALIDATION_PROTOCOL_VERSION
        ),

        "output_schema_version": (
            CELL53_OUTPUT_SCHEMA_VERSION
        ),

        "model": semantic_review_model,

        "temperature": (
            CELL53_TEMPERATURE
        ),

        "top_p": (
            CELL53_TOP_P
        ),

        "max_tokens": (
            CELL53_MAX_TOKENS
        ),
    }

    for (
        field_name,
        expected_value,
    ) in expected_values.items():

        if (
            checkpoint.get(
                field_name
            )
            != expected_value
        ):

            errors.append(
                f"{field_name.upper()}_MISMATCH"
            )

    result = checkpoint.get(
        "result"
    )

    if not isinstance(
        result,
        dict,
    ):

        errors.append(
            "CHECKPOINT_RESULT_INVALID"
        )

    elif (
        result.get(
            "candidate_id"
        )
        != request["candidate_id"]
    ):

        errors.append(
            "CHECKPOINT_RESULT_CANDIDATE_MISMATCH"
        )

    return sorted(
        set(errors)
    )


def cell53_load_checkpoint(
    request,
):

    path = (
        cell53_checkpoint_path(
            request["candidate_id"]
        )
    )

    if not path.exists():

        return (
            None,
            "MISSING",
            [],
        )

    try:

        with path.open(
            "r",
            encoding="utf-8",
        ) as handle:

            checkpoint = json.load(
                handle
            )

    except Exception as exc:

        return (
            None,
            "UNREADABLE",
            [
                f"{type(exc).__name__}: {exc}"
            ],
        )

    errors = (
        cell53_validate_checkpoint(
            checkpoint,
            request,
        )
    )

    if errors:

        return (
            checkpoint,
            "INVALID",
            errors,
        )

    return (
        checkpoint,
        "VALID",
        [],
    )


# =========================================================
# RESPONSE UTILITIES
# =========================================================

def cell53_clean_model_json(
    content,
):

    if not isinstance(
        content,
        str,
    ):

        return ""

    clean = content.strip()

    if clean.startswith("```"):

        clean = re.sub(
            r"^```(?:json)?\s*",
            "",
            clean,
            flags=re.IGNORECASE,
        )

        clean = re.sub(
            r"\s*```$",
            "",
            clean,
        )

    return clean.strip()


def cell53_is_rate_limit_exception(
    exc,
):

    error_text = str(
        exc
    ).lower()

    error_type = type(
        exc
    ).__name__.lower()

    return (
        "ratelimit" in error_type
        or
        "rate_limit" in error_type
        or
        "429" in error_text
        or
        "too many requests"
        in error_text
    )


# =========================================================
# THROTTLE
# =========================================================

cell53_request_start_lock = (
    threading.Lock()
)

cell53_last_request_start = [
    0.0
]


def cell53_wait_for_request_slot():

    with cell53_request_start_lock:

        now = time.monotonic()

        elapsed = (
            now
            - cell53_last_request_start[0]
        )

        wait_seconds = max(
            0.0,
            CELL53_REQUEST_START_SPACING_SECONDS
            - elapsed,
        )

        if wait_seconds > 0:
            time.sleep(wait_seconds)

        cell53_last_request_start[0] = (
            time.monotonic()
        )


# =========================================================
# COST
# =========================================================

cell53_has_cost_rates = (
    "INPUT_TOMAN_PER_TOKEN"
    in globals()
    and
    "OUTPUT_TOMAN_PER_TOKEN"
    in globals()
)


def cell53_estimated_cost_toman(
    input_tokens,
    output_tokens,
):

    if not cell53_has_cost_rates:
        return None

    return (
        input_tokens
        * INPUT_TOMAN_PER_TOKEN
        +
        output_tokens
        * OUTPUT_TOMAN_PER_TOKEN
    )


# =========================================================
# API FAILURE RESULT
# =========================================================

def cell53_build_api_failure_result(
    request,
    api_attempts,
    attempt_history,
):

    return {
        "candidate_id": (
            request["candidate_id"]
        ),

        "request_fingerprint": (
            request["request_fingerprint"]
        ),

        "validation_protocol_version": (
            CELL53_VALIDATION_PROTOCOL_VERSION
        ),

        "execution_status": (
            "API_FAILED"
        ),

        "api_attempts": api_attempts,

        "attempt_history": (
            attempt_history
        ),

        "finish_reason": None,

        "usage": {
            "input_tokens": 0,
            "output_tokens": 0,
            "reasoning_tokens": 0,
            "total_tokens": 0,
        },

        "estimated_cost_toman": None,

        "raw_content": None,
        "parsed_json": None,
        "parse_error": None,

        "structural_errors": [
            "API_FAILED"
        ],

        "structural_status": (
            "FAIL"
        ),

        "semantic_verdict": None,

        "validation": None,
    }


# =========================================================
# EXECUTION
# =========================================================

def cell53_execute_request(
    request,
):

    attempt_history = []
    api_attempts = 0

    for attempt_number in range(
        1,
        CELL53_MAX_ATTEMPTS + 1,
    ):

        cell53_wait_for_request_slot()

        api_attempts += 1

        try:

            response = (
                semantic_review_client
                .chat
                .completions
                .create(
                    model=semantic_review_model,

                    messages=[
                        {
                            "role": "system",
                            "content": (
                                CELL37_SYSTEM_PROMPT
                            ),
                        },
                        {
                            "role": "user",
                            "content": (
                                request["user_prompt"]
                            ),
                        },
                    ],

                    temperature=(
                        CELL53_TEMPERATURE
                    ),

                    top_p=(
                        CELL53_TOP_P
                    ),

                    max_tokens=(
                        CELL53_MAX_TOKENS
                    ),

                    timeout=(
                        CELL53_REQUEST_TIMEOUT_SECONDS
                    ),

                    extra_body={
                        "enable_thinking": False,
                    },
                )
            )

            if response is None:

                raise RuntimeError(
                    "EMPTY_API_RESPONSE"
                )

            choices = getattr(
                response,
                "choices",
                None,
            )

            if (
                not isinstance(
                    choices,
                    (list, tuple),
                )
                or len(choices) == 0
            ):

                raise RuntimeError(
                    "API_RESPONSE_CHOICES_MISSING_OR_EMPTY"
                )

            choice = choices[0]

            if choice is None:

                raise RuntimeError(
                    "API_RESPONSE_FIRST_CHOICE_IS_NONE"
                )

            message = getattr(
                choice,
                "message",
                None,
            )

            if message is None:

                raise RuntimeError(
                    "API_RESPONSE_MESSAGE_MISSING"
                )

        except Exception as exc:

            attempt_history.append({
                "attempt": attempt_number,
                "status": "API_ERROR",
                "error": (
                    f"{type(exc).__name__}: {exc}"
                ),
            })

            if (
                attempt_number
                < CELL53_MAX_ATTEMPTS
            ):

                wait_seconds = (
                    70.0
                    if cell53_is_rate_limit_exception(
                        exc
                    )
                    else (
                        CELL53_RETRY_SECONDS
                        * attempt_number
                    )
                )

                time.sleep(
                    wait_seconds
                )

                continue

            return (
                cell53_build_api_failure_result(
                    request,
                    api_attempts,
                    attempt_history,
                )
            )

        finish_reason = getattr(
            choice,
            "finish_reason",
            None,
        )

        raw_content = getattr(
            message,
            "content",
            None,
        )

        usage = getattr(
            response,
            "usage",
            None,
        )

        input_tokens = int(
            getattr(
                usage,
                "prompt_tokens",
                0,
            )
            or 0
        )

        output_tokens = int(
            getattr(
                usage,
                "completion_tokens",
                0,
            )
            or 0
        )

        total_tokens = int(
            getattr(
                usage,
                "total_tokens",
                0,
            )
            or (
                input_tokens
                + output_tokens
            )
        )

        reasoning_tokens = 0

        completion_details = getattr(
            usage,
            "completion_tokens_details",
            None,
        )

        if completion_details is not None:

            reasoning_tokens = int(
                getattr(
                    completion_details,
                    "reasoning_tokens",
                    0,
                )
                or 0
            )

        clean_content = (
            cell53_clean_model_json(
                raw_content
            )
        )

        parsed_json = None
        parse_error = None

        try:

            parsed_json = json.loads(
                clean_content
            )

        except Exception as exc:

            parse_error = (
                f"{type(exc).__name__}: {exc}"
            )

        if parsed_json is None:

            structural_errors = [
                "JSON_PARSE_FAILED"
            ]

        else:

            structural_errors = (
                cell37_validate_output(
                    parsed_json
                )
            )

        structural_status = (
            "PASS"
            if not structural_errors
            else "FAIL"
        )

        semantic_verdict = None

        if (
            structural_status
            == "PASS"
        ):

            semantic_verdict = (
                parsed_json.get(
                    "verdict"
                )
            )

        return {
            "candidate_id": (
                request["candidate_id"]
            ),

            "request_fingerprint": (
                request["request_fingerprint"]
            ),

            "validation_protocol_version": (
                CELL53_VALIDATION_PROTOCOL_VERSION
            ),

            "execution_status": (
                "COMPLETED"
            ),

            "api_attempts": api_attempts,

            "attempt_history": (
                attempt_history
            ),

            "finish_reason": (
                finish_reason
            ),

            "usage": {
                "input_tokens": (
                    input_tokens
                ),
                "output_tokens": (
                    output_tokens
                ),
                "reasoning_tokens": (
                    reasoning_tokens
                ),
                "total_tokens": (
                    total_tokens
                ),
            },

            "estimated_cost_toman": (
                cell53_estimated_cost_toman(
                    input_tokens,
                    output_tokens,
                )
            ),

            "raw_content": raw_content,
            "parsed_json": parsed_json,
            "parse_error": parse_error,

            "structural_errors": (
                structural_errors
            ),

            "structural_status": (
                structural_status
            ),

            "semantic_verdict": (
                semantic_verdict
            ),

            "validation": (
                parsed_json
                if structural_status
                == "PASS"
                else None
            ),
        }

    raise AssertionError(
        "Unexpected Cell 53 execution path."
    )


# =========================================================
# PREFLIGHT CHECKPOINT SCAN
# =========================================================

cell53_valid_before_api = {}
cell53_missing_requests = []
cell53_invalid_before_api = []


for request in cell53_requests:

    (
        checkpoint,
        status,
        errors,
    ) = cell53_load_checkpoint(
        request
    )

    candidate_id = (
        request["candidate_id"]
    )

    if status == "VALID":

        cell53_valid_before_api[
            candidate_id
        ] = checkpoint

    elif status == "MISSING":

        cell53_missing_requests.append(
            request
        )

    else:

        cell53_invalid_before_api.append({
            "candidate_id": candidate_id,
            "status": status,
            "errors": errors,
        })


assert not cell53_invalid_before_api, (
    "Invalid/unreadable Cell 53 checkpoints detected. "
    "Refusing automatic paid replacement. "
    f"Examples: {cell53_invalid_before_api[:5]}"
)


# =========================================================
# PREFLIGHT OUTPUT
# =========================================================

print(
    "=== GOLDEN V2 TARGETED QA REPAIR "
    "SEMANTIC REVALIDATION PREFLIGHT ==="
)

print()

print(
    "Full population:",
    3813,
)

print(
    "Previously accepted semantic PASS preserved:",
    len(
        cell53_preserved_semantic_results
    ),
)

print(
    "Repair revalidation population:",
    len(
        cell53_requests
    ),
)

print(
    "Exact valid revalidation checkpoints:",
    len(
        cell53_valid_before_api
    ),
)

print(
    "Missing revalidation checkpoints:",
    len(
        cell53_missing_requests
    ),
)

print(
    "Model:",
    semantic_review_model,
)

print(
    "Validation protocol:",
    CELL53_VALIDATION_PROTOCOL_VERSION,
)

print(
    "Max tokens:",
    CELL53_MAX_TOKENS,
)

print(
    "Parallel workers:",
    CELL53_MAX_WORKERS,
)

print(
    "Request spacing seconds:",
    CELL53_REQUEST_START_SPACING_SECONDS,
)

print(
    "API allowed for missing exact checkpoints:",
    CELL53_ALLOW_API_FOR_MISSING,
)

print(
    "Previously accepted 3776 QA sent to API:",
    0,
)

print()


if (
    cell53_missing_requests
    and
    not CELL53_ALLOW_API_FOR_MISSING
):

    raise AssertionError(
        "Cell 53 has missing checkpoints "
        "but API execution is disabled."
    )


# =========================================================
# EXECUTE + CHECKPOINT
# =========================================================

cell53_api_candidates_this_execution = 0
cell53_api_attempts_this_execution = 0
cell53_checkpoints_created_this_execution = 0


def cell53_execute_and_checkpoint(
    request,
):

    candidate_id = (
        request["candidate_id"]
    )

    result = (
        cell53_execute_request(
            request
        )
    )

    checkpoint = (
        cell53_build_checkpoint(
            request,
            result,
        )
    )

    cell53_atomic_write_json(
        cell53_checkpoint_path(
            candidate_id
        ),
        checkpoint,
    )

    (
        verified_checkpoint,
        verified_status,
        verified_errors,
    ) = cell53_load_checkpoint(
        request
    )

    if verified_status != "VALID":

        raise AssertionError(
            "Cell 53 checkpoint verification "
            f"failed for {candidate_id}: "
            f"{verified_errors}"
        )

    return (
        candidate_id,
        verified_checkpoint["result"],
    )


if cell53_missing_requests:

    completed_count = 0

    executor = ThreadPoolExecutor(
        max_workers=CELL53_MAX_WORKERS
    )

    future_to_request = {}

    try:

        future_to_request = {
            executor.submit(
                cell53_execute_and_checkpoint,
                request,
            ): request

            for request
            in cell53_missing_requests
        }

        for future in as_completed(
            future_to_request
        ):

            request = (
                future_to_request[future]
            )

            candidate_id = (
                request["candidate_id"]
            )

            try:

                (
                    resolved_candidate_id,
                    result,
                ) = future.result()

            except Exception as exc:

                completed_count += 1

                print(
                    f"[{completed_count:02d}/"
                    f"{len(cell53_missing_requests):02d}]",
                    candidate_id,
                    "| WORKER_EXCEPTION =",
                    f"{type(exc).__name__}: {exc}",
                    flush=True,
                )

                continue

            assert (
                resolved_candidate_id
                == candidate_id
            )

            completed_count += 1

            cell53_api_candidates_this_execution += 1

            cell53_api_attempts_this_execution += int(
                result.get(
                    "api_attempts",
                    0,
                )
                or 0
            )

            cell53_checkpoints_created_this_execution += 1

            print(
                f"[{completed_count:02d}/"
                f"{len(cell53_missing_requests):02d}]",
                candidate_id,
                "| structural=",
                result.get(
                    "structural_status"
                ),
                "| semantic=",
                result.get(
                    "semantic_verdict"
                ),
                "| finish=",
                result.get(
                    "finish_reason"
                ),
                flush=True,
            )

    except KeyboardInterrupt:

        print(
            "\nCell 53 interrupted by user. "
            "Completed checkpoints remain preserved.",
            flush=True,
        )

        for future in future_to_request:

            if not future.done():
                future.cancel()

        raise

    finally:

        executor.shutdown(
            wait=False,
            cancel_futures=True,
        )


# =========================================================
# FINAL CHECKPOINT RESCAN
# =========================================================

cell53_revalidation_results = []

cell53_valid_checkpoint_count = 0
cell53_missing_checkpoint_count = 0
cell53_invalid_checkpoint_count = 0


for request in cell53_requests:

    (
        checkpoint,
        status,
        errors,
    ) = cell53_load_checkpoint(
        request
    )

    if status == "VALID":

        cell53_valid_checkpoint_count += 1

        cell53_revalidation_results.append(
            checkpoint["result"]
        )

    elif status == "MISSING":

        cell53_missing_checkpoint_count += 1

    else:

        cell53_invalid_checkpoint_count += 1


# =========================================================
# REVALIDATION DISTRIBUTIONS
# =========================================================

cell53_structural_distribution = Counter(
    result.get("structural_status")
    for result
    in cell53_revalidation_results
)

cell53_execution_distribution = Counter(
    result.get("execution_status")
    for result
    in cell53_revalidation_results
)

cell53_finish_distribution = Counter(
    result.get("finish_reason")
    for result
    in cell53_revalidation_results
)

cell53_semantic_distribution = Counter(
    result.get("semantic_verdict")
    for result
    in cell53_revalidation_results
    if result.get("structural_status")
    == "PASS"
)


# =========================================================
# MERGE WITH PRESERVED 3776
# =========================================================

cell53_revalidation_by_candidate_id = {
    result["candidate_id"]: result
    for result
    in cell53_revalidation_results
}

cell53_final_semantic_results = []


for skeleton in cell25_candidate_skeletons:

    candidate_id = (
        skeleton["candidate_id"]
    )

    if candidate_id in (
        cell53_revalidation_by_candidate_id
    ):

        cell53_final_semantic_results.append(
            cell53_revalidation_by_candidate_id[
                candidate_id
            ]
        )

    else:

        cell53_final_semantic_results.append(
            cell53_previous_semantic_by_candidate_id[
                candidate_id
            ]
        )


assert len(
    cell53_final_semantic_results
) == 3813


# =========================================================
# FINAL 3813 STATE
# =========================================================

cell53_final_structural_distribution = Counter(
    result.get("structural_status")
    for result
    in cell53_final_semantic_results
)

cell53_final_execution_distribution = Counter(
    result.get("execution_status")
    for result
    in cell53_final_semantic_results
)

cell53_final_semantic_distribution = Counter(
    result.get("semantic_verdict")
    for result
    in cell53_final_semantic_results
    if result.get("structural_status")
    == "PASS"
)


cell53_final_structural_fail_count = (
    cell53_final_structural_distribution.get(
        "FAIL",
        0,
    )
)

cell53_final_api_failure_count = (
    cell53_final_execution_distribution.get(
        "API_FAILED",
        0,
    )
)

cell53_final_semantic_pass_count = (
    cell53_final_semantic_distribution.get(
        "PASS",
        0,
    )
)

cell53_final_semantic_fail_count = (
    cell53_final_semantic_distribution.get(
        "FAIL",
        0,
    )
)


cell53_remaining_semantic_fail_results = [
    result
    for result
    in cell53_final_semantic_results
    if (
        result.get("structural_status")
        == "PASS"
        and
        result.get("semantic_verdict")
        == "FAIL"
    )
]

cell53_remaining_semantic_fail_candidate_ids = [
    result["candidate_id"]
    for result
    in cell53_remaining_semantic_fail_results
]


# =========================================================
# COST
# =========================================================

cell53_known_costs = [
    result.get(
        "estimated_cost_toman"
    )
    for result
    in cell53_revalidation_results
    if isinstance(
        result.get(
            "estimated_cost_toman"
        ),
        (int, float),
    )
]

cell53_estimated_cost_toman = (
    sum(cell53_known_costs)
    if cell53_known_costs
    else None
)


# =========================================================
# RUNNER HEALTH
# =========================================================

cell53_checks = {
    "full_population_3813": (
        len(
            cell53_final_semantic_results
        )
        == 3813
    ),

    "preserved_semantic_pass_3776": (
        len(
            cell53_preserved_semantic_results
        )
        == 3776
    ),

    "revalidation_population_37": (
        len(cell53_requests)
        == 37
    ),

    "all_revalidation_checkpoints_present": (
        cell53_valid_checkpoint_count
        == 37
    ),

    "zero_missing_checkpoints": (
        cell53_missing_checkpoint_count
        == 0
    ),

    "zero_invalid_checkpoints": (
        cell53_invalid_checkpoint_count
        == 0
    ),

    "candidate_identity_exact": (
        len({
            result["candidate_id"]
            for result
            in cell53_final_semantic_results
        })
        == 3813
        and
        set(
            result["candidate_id"]
            for result
            in cell53_final_semantic_results
        )
        ==
        set(
            cell53_previous_semantic_by_candidate_id
        )
    ),

    "previously_accepted_3776_not_revalidated": True,
}


cell53_runner_status = (
    "PASS"
    if all(
        cell53_checks.values()
    )
    else "FAIL"
)


if cell53_runner_status != "PASS":

    cell53_next_readiness = (
        "BLOCKED_CELL53_RUNNER_REPAIR_REQUIRED"
    )

elif cell53_final_api_failure_count > 0:

    cell53_next_readiness = (
        "READY_FOR_CELL53_TECHNICAL_RECOVERY"
    )

elif cell53_final_structural_fail_count > 0:

    cell53_next_readiness = (
        "READY_FOR_CELL53_STRUCTURAL_RECOVERY"
    )

elif cell53_final_semantic_fail_count > 0:

    cell53_next_readiness = (
        "READY_FOR_RESIDUAL_SEMANTIC_FAILURE_ADJUDICATION"
    )

elif cell53_final_semantic_pass_count == 3813:

    cell53_next_readiness = (
        "READY_FOR_GOLDEN_V2_FINAL_RECORD_CONSTRUCTION"
    )

else:

    cell53_next_readiness = (
        "READY_FOR_CELL53_RESULT_ANALYSIS"
    )


# =========================================================
# OUTPUT
# =========================================================

print()

print(
    "=== GOLDEN V2 TARGETED QA REPAIR "
    "SEMANTIC REVALIDATION RESULT ==="
)

print()

print(
    "Revalidation population:",
    len(cell53_requests),
)

print(
    "Valid checkpoints:",
    cell53_valid_checkpoint_count,
)

print(
    "Missing checkpoints:",
    cell53_missing_checkpoint_count,
)

print(
    "Invalid checkpoints:",
    cell53_invalid_checkpoint_count,
)

print()

print(
    "API candidates this execution:",
    cell53_api_candidates_this_execution,
)

print(
    "API attempts this execution:",
    cell53_api_attempts_this_execution,
)

print(
    "Checkpoints created this execution:",
    cell53_checkpoints_created_this_execution,
)

print()

print(
    "Repair revalidation structural distribution:",
    dict(
        cell53_structural_distribution
    ),
)

print(
    "Repair revalidation semantic distribution:",
    dict(
        cell53_semantic_distribution
    ),
)

print(
    "Finish reason distribution:",
    dict(
        cell53_finish_distribution
    ),
)

print(
    "Estimated revalidation cost (toman):",
    cell53_estimated_cost_toman,
)

print()

print(
    "=== FINAL 3813 SEMANTIC STATE ==="
)

print(
    "Structural distribution:",
    dict(
        cell53_final_structural_distribution
    ),
)

print(
    "API failures:",
    cell53_final_api_failure_count,
)

print(
    "Semantic distribution:",
    dict(
        cell53_final_semantic_distribution
    ),
)

print(
    "Semantic PASS:",
    cell53_final_semantic_pass_count,
)

print(
    "Semantic FAIL:",
    cell53_final_semantic_fail_count,
)

print()

print(
    "Remaining semantic FAIL candidate IDs:"
)

for candidate_id in (
    cell53_remaining_semantic_fail_candidate_ids
):

    print(
        " -",
        candidate_id,
    )

print()

print("Runner checks:")

for (
    check_name,
    check_value,
) in cell53_checks.items():

    print(
        " -",
        check_name,
        ":",
        (
            "PASS"
            if check_value
            else "FAIL"
        ),
    )

print()

print(
    "Runner health:",
    cell53_runner_status,
)

print(
    "Next readiness:",
    cell53_next_readiness,
)

print(
    "CELL 53 CHECKPOINTED RUNNER:",
    cell53_runner_status,
)


assert cell53_runner_status == "PASS", (
    "Cell 53 checkpointed runner health failed."
)
````

````python
# Code Cell 54 — Zero-API Adjudication of the Single Residual Semantic Failure

from collections import Counter
import json


# =========================================================
# CONFIG
# =========================================================

CELL54_VERSION = (
    "golden_v2_single_residual_semantic_failure_adjudication_v1"
)


# =========================================================
# REQUIRED STATE
# =========================================================

cell54_required_names = [
    "cell25_candidate_skeletons",
    "cell47_final_generation_results",
    "cell50_failure_adjudications",
    "cell52_final_repair_results",
    "cell53_final_semantic_results",
    "cell53_remaining_semantic_fail_candidate_ids",
    "CELL37_BOOLEAN_FIELDS",
]

cell54_missing_state = [
    name
    for name in cell54_required_names
    if name not in globals()
]

assert not cell54_missing_state, (
    "Missing required Cell 54 state: "
    + ", ".join(cell54_missing_state)
)


# =========================================================
# EXACT FINAL STATE FROM CELL 53
# =========================================================

assert len(
    cell53_final_semantic_results
) == 3813

assert len(
    cell53_remaining_semantic_fail_candidate_ids
) == 1, (
    "Expected exactly one residual Semantic FAIL."
)


cell54_candidate_id = (
    cell53_remaining_semantic_fail_candidate_ids[0]
)

assert (
    cell54_candidate_id
    ==
    "golden_v2_candidate_f4b2f9e2cb3874670125b227"
), (
    "Unexpected residual Semantic FAIL candidate."
)


# =========================================================
# LOOKUPS
# =========================================================

cell54_skeleton_by_candidate_id = {
    record["candidate_id"]: record
    for record in cell25_candidate_skeletons
}

cell54_original_generation_by_candidate_id = {
    result["candidate_id"]: result
    for result in cell47_final_generation_results
}

cell54_original_adjudication_by_candidate_id = {
    record["candidate_id"]: record
    for record in cell50_failure_adjudications
}

cell54_repair_by_candidate_id = {
    result["candidate_id"]: result
    for result in cell52_final_repair_results
}

cell54_final_semantic_by_candidate_id = {
    result["candidate_id"]: result
    for result in cell53_final_semantic_results
}


assert cell54_candidate_id in (
    cell54_skeleton_by_candidate_id
)

assert cell54_candidate_id in (
    cell54_original_generation_by_candidate_id
)

assert cell54_candidate_id in (
    cell54_original_adjudication_by_candidate_id
)

assert cell54_candidate_id in (
    cell54_repair_by_candidate_id
)

assert cell54_candidate_id in (
    cell54_final_semantic_by_candidate_id
)


# =========================================================
# SOURCE RECORDS
# =========================================================

cell54_skeleton = (
    cell54_skeleton_by_candidate_id[
        cell54_candidate_id
    ]
)

cell54_original_generation = (
    cell54_original_generation_by_candidate_id[
        cell54_candidate_id
    ]
)

cell54_original_adjudication = (
    cell54_original_adjudication_by_candidate_id[
        cell54_candidate_id
    ]
)

cell54_repair_result = (
    cell54_repair_by_candidate_id[
        cell54_candidate_id
    ]
)

cell54_final_semantic_result = (
    cell54_final_semantic_by_candidate_id[
        cell54_candidate_id
    ]
)


# =========================================================
# BASIC VALIDITY
# =========================================================

assert (
    cell54_repair_result.get(
        "structural_status"
    )
    == "PASS"
)

assert (
    cell54_final_semantic_result.get(
        "structural_status"
    )
    == "PASS"
)

assert (
    cell54_final_semantic_result.get(
        "semantic_verdict"
    )
    == "FAIL"
)


cell54_original_qa = (
    cell54_original_generation.get(
        "parsed_json"
    )
    or {}
)

cell54_repaired_qa = (
    cell54_repair_result.get(
        "parsed_json"
    )
    or {}
)

cell54_validation = (
    cell54_final_semantic_result.get(
        "validation"
    )
    or {}
)


assert isinstance(
    cell54_original_qa,
    dict,
)

assert isinstance(
    cell54_repaired_qa,
    dict,
)

assert isinstance(
    cell54_validation,
    dict,
)


# =========================================================
# FAILED DIMENSIONS
# =========================================================

cell54_failed_dimensions = [
    dimension
    for dimension
    in CELL37_BOOLEAN_FIELDS
    if (
        cell54_validation.get(
            dimension
        )
        is False
    )
]


cell54_failure_reasons = [
    str(reason).strip()
    for reason
    in (
        cell54_validation.get(
            "failure_reasons",
            []
        )
        or []
    )
    if str(reason).strip()
]


assert cell54_failed_dimensions, (
    "Semantic FAIL has no failed validation dimension."
)


# =========================================================
# FAILURE CLASSIFICATION
# =========================================================

def cell54_classify_failure(
    failed_dimensions,
):

    failed = set(
        failed_dimensions
    )

    classes = []

    if (
        "question_targets_claim"
        in failed
        or
        "answer_question_alignment"
        in failed
    ):
        classes.append(
            "QUESTION_TARGET_OR_ALIGNMENT_DRIFT"
        )

    if (
        "answer_recovers_claim"
        in failed
    ):
        classes.append(
            "ANSWER_CLAIM_RECOVERY_DRIFT"
        )

    if (
        "scope_fidelity"
        in failed
    ):
        classes.append(
            "SCOPE_FIDELITY_DRIFT"
        )

    if (
        "modality_quantifier_fidelity"
        in failed
    ):
        classes.append(
            "MODALITY_OR_QUANTIFIER_DRIFT"
        )

    if (
        "standalone_question"
        in failed
    ):
        classes.append(
            "QUESTION_NOT_STANDALONE"
        )

    if (
        "no_neighbor_fact_drift"
        in failed
    ):
        classes.append(
            "NEIGHBOR_FACT_DRIFT"
        )

    if (
        "no_external_knowledge"
        in failed
    ):
        classes.append(
            "EXTERNAL_KNOWLEDGE_DRIFT"
        )

    if not classes:
        classes.append(
            "UNCLASSIFIED_SEMANTIC_FAILURE"
        )

    return classes


cell54_failure_classes = (
    cell54_classify_failure(
        cell54_failed_dimensions
    )
)


# =========================================================
# BUILD EXACT FINAL ADJUDICATION RECORD
# =========================================================

cell54_residual_failure_adjudication = {
    "adjudication_version": (
        CELL54_VERSION
    ),

    "candidate_id": (
        cell54_candidate_id
    ),

    "claim_id": (
        cell54_skeleton[
            "claim_id"
        ]
    ),

    "claim_text_fa": (
        cell54_skeleton[
            "claim_text_fa"
        ]
    ),

    "evidence_ids": list(
        cell54_skeleton[
            "evidence_ids"
        ]
    ),

    "evidence_quotes": list(
        cell54_skeleton[
            "evidence_quotes"
        ]
    ),

    "original_qa": {
        "question": (
            cell54_original_qa.get(
                "question"
            )
        ),

        "expected_answer": (
            cell54_original_qa.get(
                "expected_answer"
            )
        ),

        "question_intent": (
            cell54_original_qa.get(
                "question_intent"
            )
        ),
    },

    "first_failure": {
        "failed_dimensions": list(
            cell54_original_adjudication[
                "failed_dimensions"
            ]
        ),

        "failure_classes": list(
            cell54_original_adjudication[
                "failure_classes"
            ]
        ),

        "failure_reasons": list(
            cell54_original_adjudication[
                "failure_reasons"
            ]
        ),
    },

    "repaired_qa": {
        "question": (
            cell54_repaired_qa.get(
                "question"
            )
        ),

        "expected_answer": (
            cell54_repaired_qa.get(
                "expected_answer"
            )
        ),

        "question_intent": (
            cell54_repaired_qa.get(
                "question_intent"
            )
        ),
    },

    "revalidation": {
        "verdict": (
            cell54_final_semantic_result.get(
                "semantic_verdict"
            )
        ),

        "failed_dimensions": (
            cell54_failed_dimensions
        ),

        "failure_classes": (
            cell54_failure_classes
        ),

        "failure_reasons": (
            cell54_failure_reasons
        ),

        "full_validation": (
            cell54_validation
        ),
    },

    "repair_target": (
        "GENERATED_QA"
    ),

    "preserve_claim": True,
    "preserve_evidence": True,

    "recommended_next_action": (
        "FINAL_SINGLE_CANDIDATE_QA_REPAIR"
    ),
}


# =========================================================
# FINAL POPULATION SAFETY
# =========================================================

cell54_semantic_pass_count = sum(
    1
    for result
    in cell53_final_semantic_results
    if (
        result.get(
            "structural_status"
        )
        == "PASS"
        and
        result.get(
            "semantic_verdict"
        )
        == "PASS"
    )
)

cell54_semantic_fail_count = sum(
    1
    for result
    in cell53_final_semantic_results
    if (
        result.get(
            "structural_status"
        )
        == "PASS"
        and
        result.get(
            "semantic_verdict"
        )
        == "FAIL"
    )
)


cell54_checks = {
    "full_population_3813": (
        len(
            cell53_final_semantic_results
        )
        == 3813
    ),

    "semantic_pass_3812": (
        cell54_semantic_pass_count
        == 3812
    ),

    "semantic_fail_1": (
        cell54_semantic_fail_count
        == 1
    ),

    "exact_residual_candidate": (
        cell54_candidate_id
        ==
        "golden_v2_candidate_f4b2f9e2cb3874670125b227"
    ),

    "residual_repair_structural_pass": (
        cell54_repair_result.get(
            "structural_status"
        )
        == "PASS"
    ),

    "residual_revalidation_structural_pass": (
        cell54_final_semantic_result.get(
            "structural_status"
        )
        == "PASS"
    ),

    "residual_semantic_fail": (
        cell54_final_semantic_result.get(
            "semantic_verdict"
        )
        == "FAIL"
    ),

    "failed_dimension_present": (
        len(
            cell54_failed_dimensions
        )
        >= 1
    ),

    "repair_target_generated_qa": (
        cell54_residual_failure_adjudication[
            "repair_target"
        ]
        == "GENERATED_QA"
    ),

    "claim_preserved": (
        cell54_residual_failure_adjudication[
            "preserve_claim"
        ]
        is True
    ),

    "evidence_preserved": (
        cell54_residual_failure_adjudication[
            "preserve_evidence"
        ]
        is True
    ),

    "zero_api_calls": True,
    "zero_file_writes": True,
}


cell54_status = (
    "PASS"
    if all(
        cell54_checks.values()
    )
    else "FAIL"
)


cell54_next_readiness = (
    "READY_FOR_FINAL_SINGLE_CANDIDATE_QA_REPAIR"
    if cell54_status == "PASS"
    else
    "BLOCKED_CELL54_REPAIR_REQUIRED"
)


# =========================================================
# OUTPUT
# =========================================================

print(
    "=== GOLDEN V2 SINGLE RESIDUAL "
    "SEMANTIC FAILURE ADJUDICATION ==="
)

print()

print(
    "Full population:",
    len(
        cell53_final_semantic_results
    ),
)

print(
    "Semantic PASS preserved:",
    cell54_semantic_pass_count,
)

print(
    "Residual Semantic FAIL:",
    cell54_semantic_fail_count,
)

print()

print(
    "Candidate:",
    cell54_candidate_id,
)

print(
    "Claim:",
    cell54_residual_failure_adjudication[
        "claim_text_fa"
    ],
)

print()

print(
    "Original Question:",
    cell54_residual_failure_adjudication[
        "original_qa"
    ][
        "question"
    ],
)

print(
    "Original Expected Answer:",
    cell54_residual_failure_adjudication[
        "original_qa"
    ][
        "expected_answer"
    ],
)

print()

print(
    "Repaired Question:",
    cell54_residual_failure_adjudication[
        "repaired_qa"
    ][
        "question"
    ],
)

print(
    "Repaired Expected Answer:",
    cell54_residual_failure_adjudication[
        "repaired_qa"
    ][
        "expected_answer"
    ],
)

print()

print(
    "Original failed dimensions:",
    cell54_residual_failure_adjudication[
        "first_failure"
    ][
        "failed_dimensions"
    ],
)

print(
    "Residual failed dimensions:",
    cell54_failed_dimensions,
)

print(
    "Residual failure classes:",
    cell54_failure_classes,
)

print(
    "Residual failure reasons:",
    cell54_failure_reasons,
)

print()

print("Checks:")

for (
    check_name,
    check_value,
) in cell54_checks.items():

    print(
        " -",
        check_name,
        ":",
        (
            "PASS"
            if check_value
            else "FAIL"
        ),
    )

print()

print(
    "CELL 54 STATUS:",
    cell54_status,
)

print(
    "Next readiness:",
    cell54_next_readiness,
)


assert cell54_status == "PASS", (
    "Cell 54 adjudication failed."
)
````

````python
# Code Cell 55 — Final Single-Candidate QA Repair

from pathlib import Path
import hashlib
import json
import os
import re
import tempfile
import time


# =========================================================
# CONFIG
# =========================================================

CELL55_RUNNER_VERSION = (
    "golden_v2_final_single_candidate_qa_repair_v1"
)

CELL55_PROMPT_VERSION = (
    "golden_v2_final_single_candidate_qa_repair_prompt_v1"
)

CELL55_MAX_TOKENS = 3072
CELL55_TEMPERATURE = 0.7
CELL55_TOP_P = 0.8

CELL55_MAX_ATTEMPTS = 2
CELL55_RETRY_SECONDS = 2.0
CELL55_REQUEST_TIMEOUT_SECONDS = 180.0

# Before final Restart -> Run All this MUST become False.
CELL55_ALLOW_API_FOR_MISSING = False


# =========================================================
# REQUIRED STATE
# =========================================================

cell55_required_names = [
    "project_root",
    "semantic_review_client",
    "semantic_review_model",
    "GOLDEN_V2_GENERATION_SYSTEM_PROMPT_V4",
    "GENERATION_V4_VERSION",
    "GENERATION_V4_OUTPUT_SCHEMA_VERSION",
    "cell41_validate_generation_output_v4",
    "cell54_residual_failure_adjudication",
]

cell55_missing_state = [
    name
    for name in cell55_required_names
    if name not in globals()
]

assert not cell55_missing_state, (
    "Missing required Cell 55 state: "
    + ", ".join(cell55_missing_state)
)


# =========================================================
# EXACT SINGLE TARGET
# =========================================================

CELL55_CANDIDATE_ID = (
    "golden_v2_candidate_f4b2f9e2cb3874670125b227"
)

assert (
    cell54_residual_failure_adjudication[
        "candidate_id"
    ]
    == CELL55_CANDIDATE_ID
)

assert (
    cell54_residual_failure_adjudication[
        "repair_target"
    ]
    == "GENERATED_QA"
)

assert (
    cell54_residual_failure_adjudication[
        "preserve_claim"
    ]
    is True
)

assert (
    cell54_residual_failure_adjudication[
        "preserve_evidence"
    ]
    is True
)


# =========================================================
# HASH UTILITIES
# =========================================================

def cell55_canonical_json(value):

    return json.dumps(
        value,
        ensure_ascii=False,
        sort_keys=True,
        separators=(",", ":"),
    )


def cell55_sha256_text(text):

    return hashlib.sha256(
        text.encode("utf-8")
    ).hexdigest()


# =========================================================
# FINAL REPAIR PACKET
# =========================================================

cell55_repair_packet = {
    "candidate_id": (
        CELL55_CANDIDATE_ID
    ),

    "claim": (
        cell54_residual_failure_adjudication[
            "claim_text_fa"
        ]
    ),

    "evidence_ids": list(
        cell54_residual_failure_adjudication[
            "evidence_ids"
        ]
    ),

    "evidence_quotes": list(
        cell54_residual_failure_adjudication[
            "evidence_quotes"
        ]
    ),

    "previous_repaired_qa": (
        cell54_residual_failure_adjudication[
            "repaired_qa"
        ]
    ),

    "residual_failed_dimensions": (
        cell54_residual_failure_adjudication[
            "revalidation"
        ][
            "failed_dimensions"
        ]
    ),

    "residual_failure_reasons": (
        cell54_residual_failure_adjudication[
            "revalidation"
        ][
            "failure_reasons"
        ]
    ),
}


assert (
    cell55_repair_packet[
        "residual_failed_dimensions"
    ]
    == ["question_targets_claim"]
), (
    "Unexpected residual failure dimensions."
)


# =========================================================
# TARGETED PROMPT
# =========================================================

CELL55_USER_PROMPT = (
    "Repair exactly one generated QA pair.\n\n"

    "The Claim and Evidence are already validated and MUST NOT "
    "be changed, expanded, rejected, reinterpreted or repaired.\n\n"

    "The previous repair failed ONLY because the Question was too "
    "general and did not directly target the Claim's specific topic.\n\n"

    "REPAIR REQUIREMENTS:\n"
    "1. Generate a standalone Persian Question that explicitly "
    "mentions the relevant subject of the Claim: learning more "
    "about credit cards / کارت‌های اعتباری.\n"
    "2. The Question must directly test the Claim's assertion that "
    "knowing more about credit cards is good.\n"
    "3. Do not use the generic Question «چه چیزی خوب است؟» or any "
    "equally context-dependent wording.\n"
    "4. Do not introduce WHY, cause, mechanism, benefit, consequence "
    "or any information not asserted by the Claim.\n"
    "5. Preserve the evaluative meaning «خوب است».\n"
    "6. Expected Answer must directly answer the generated Question "
    "and faithfully recover the Claim.\n"
    "7. Do not introduce external knowledge.\n"
    "8. Return ONLY one JSON object with exactly these fields:\n"
    "{\n"
    '  "question": "...",\n'
    '  "expected_answer": "...",\n'
    '  "question_intent": "..."\n'
    "}\n\n"

    "REPAIR PACKET:\n"
    + json.dumps(
        cell55_repair_packet,
        ensure_ascii=False,
        sort_keys=True,
        indent=2,
    )
)


# =========================================================
# REQUEST CONTRACT
# =========================================================

cell55_request_contract = {
    "runner_version": (
        CELL55_RUNNER_VERSION
    ),

    "generation_protocol_version": (
        GENERATION_V4_VERSION
    ),

    "prompt_version": (
        CELL55_PROMPT_VERSION
    ),

    "output_schema_version": (
        GENERATION_V4_OUTPUT_SCHEMA_VERSION
    ),

    "candidate_id": (
        CELL55_CANDIDATE_ID
    ),

    "repair_packet_sha256": (
        cell55_sha256_text(
            cell55_canonical_json(
                cell55_repair_packet
            )
        )
    ),

    "model": (
        semantic_review_model
    ),

    "system_prompt": (
        GOLDEN_V2_GENERATION_SYSTEM_PROMPT_V4
    ),

    "user_prompt": (
        CELL55_USER_PROMPT
    ),

    "temperature": (
        CELL55_TEMPERATURE
    ),

    "top_p": (
        CELL55_TOP_P
    ),

    "max_tokens": (
        CELL55_MAX_TOKENS
    ),

    "enable_thinking": False,
}


CELL55_REQUEST_FINGERPRINT = (
    cell55_sha256_text(
        cell55_canonical_json(
            cell55_request_contract
        )
    )
)


# =========================================================
# CHECKPOINT
# =========================================================

cell55_checkpoint_dir = (
    Path(project_root)
    / "artifacts"
    / "phase2"
    / "golden_dataset_v2"
    / "work"
    / "final_single_candidate_qa_repair_v1"
)

cell55_checkpoint_dir.mkdir(
    parents=True,
    exist_ok=True,
)

cell55_checkpoint_path = (
    cell55_checkpoint_dir
    / f"{CELL55_CANDIDATE_ID}.json"
)


def cell55_atomic_write_json(
    path,
    payload,
):

    path = Path(path)

    content = (
        json.dumps(
            payload,
            ensure_ascii=False,
            sort_keys=True,
            indent=2,
        )
        + "\n"
    )

    fd, temp_name = tempfile.mkstemp(
        prefix=path.name + ".",
        suffix=".tmp",
        dir=str(path.parent),
    )

    try:

        with os.fdopen(
            fd,
            "w",
            encoding="utf-8",
        ) as handle:

            handle.write(content)
            handle.flush()
            os.fsync(handle.fileno())

        os.replace(
            temp_name,
            path,
        )

    except Exception:

        try:
            os.unlink(temp_name)
        except FileNotFoundError:
            pass

        raise


def cell55_build_checkpoint(result):

    checkpoint = {
        "checkpoint_version": (
            CELL55_RUNNER_VERSION
        ),

        "candidate_id": (
            CELL55_CANDIDATE_ID
        ),

        "request_fingerprint": (
            CELL55_REQUEST_FINGERPRINT
        ),

        "repair_packet_sha256": (
            cell55_request_contract[
                "repair_packet_sha256"
            ]
        ),

        "generation_protocol_version": (
            GENERATION_V4_VERSION
        ),

        "prompt_version": (
            CELL55_PROMPT_VERSION
        ),

        "output_schema_version": (
            GENERATION_V4_OUTPUT_SCHEMA_VERSION
        ),

        "model": (
            semantic_review_model
        ),

        "temperature": (
            CELL55_TEMPERATURE
        ),

        "top_p": (
            CELL55_TOP_P
        ),

        "max_tokens": (
            CELL55_MAX_TOKENS
        ),

        "result": result,
    }

    checkpoint[
        "checkpoint_payload_sha256"
    ] = cell55_sha256_text(
        cell55_canonical_json(
            checkpoint
        )
    )

    return checkpoint


def cell55_validate_checkpoint(
    checkpoint,
):

    errors = []

    if not isinstance(
        checkpoint,
        dict,
    ):

        return [
            "CHECKPOINT_NOT_OBJECT"
        ]

    stored_sha = checkpoint.get(
        "checkpoint_payload_sha256"
    )

    payload_without_sha = {
        key: value
        for key, value
        in checkpoint.items()
        if key
        != "checkpoint_payload_sha256"
    }

    recalculated_sha = (
        cell55_sha256_text(
            cell55_canonical_json(
                payload_without_sha
            )
        )
    )

    if stored_sha != recalculated_sha:

        errors.append(
            "CHECKPOINT_HASH_MISMATCH"
        )

    expected = {
        "checkpoint_version": (
            CELL55_RUNNER_VERSION
        ),

        "candidate_id": (
            CELL55_CANDIDATE_ID
        ),

        "request_fingerprint": (
            CELL55_REQUEST_FINGERPRINT
        ),

        "repair_packet_sha256": (
            cell55_request_contract[
                "repair_packet_sha256"
            ]
        ),

        "generation_protocol_version": (
            GENERATION_V4_VERSION
        ),

        "prompt_version": (
            CELL55_PROMPT_VERSION
        ),

        "output_schema_version": (
            GENERATION_V4_OUTPUT_SCHEMA_VERSION
        ),

        "model": (
            semantic_review_model
        ),

        "temperature": (
            CELL55_TEMPERATURE
        ),

        "top_p": (
            CELL55_TOP_P
        ),

        "max_tokens": (
            CELL55_MAX_TOKENS
        ),
    }

    for field_name, expected_value in (
        expected.items()
    ):

        if (
            checkpoint.get(
                field_name
            )
            != expected_value
        ):

            errors.append(
                f"{field_name.upper()}_MISMATCH"
            )

    result = checkpoint.get(
        "result"
    )

    if not isinstance(
        result,
        dict,
    ):

        errors.append(
            "RESULT_INVALID"
        )

    elif (
        result.get(
            "candidate_id"
        )
        != CELL55_CANDIDATE_ID
    ):

        errors.append(
            "RESULT_CANDIDATE_MISMATCH"
        )

    return sorted(set(errors))


def cell55_load_checkpoint():

    if not cell55_checkpoint_path.exists():

        return (
            None,
            "MISSING",
            [],
        )

    try:

        with cell55_checkpoint_path.open(
            "r",
            encoding="utf-8",
        ) as handle:

            checkpoint = json.load(
                handle
            )

    except Exception as exc:

        return (
            None,
            "UNREADABLE",
            [
                f"{type(exc).__name__}: {exc}"
            ],
        )

    errors = (
        cell55_validate_checkpoint(
            checkpoint
        )
    )

    if errors:

        return (
            checkpoint,
            "INVALID",
            errors,
        )

    return (
        checkpoint,
        "VALID",
        [],
    )


# =========================================================
# RESPONSE UTILITIES
# =========================================================

def cell55_clean_json(content):

    if not isinstance(content, str):
        return ""

    clean = content.strip()

    if clean.startswith("```"):

        clean = re.sub(
            r"^```(?:json)?\s*",
            "",
            clean,
            flags=re.IGNORECASE,
        )

        clean = re.sub(
            r"\s*```$",
            "",
            clean,
        )

    return clean.strip()


# =========================================================
# API EXECUTION
# =========================================================

def cell55_execute_api():

    attempt_history = []

    for attempt_number in range(
        1,
        CELL55_MAX_ATTEMPTS + 1,
    ):

        try:

            response = (
                semantic_review_client
                .chat
                .completions
                .create(
                    model=semantic_review_model,

                    messages=[
                        {
                            "role": "system",
                            "content": (
                                GOLDEN_V2_GENERATION_SYSTEM_PROMPT_V4
                            ),
                        },
                        {
                            "role": "user",
                            "content": (
                                CELL55_USER_PROMPT
                            ),
                        },
                    ],

                    temperature=(
                        CELL55_TEMPERATURE
                    ),

                    top_p=(
                        CELL55_TOP_P
                    ),

                    max_tokens=(
                        CELL55_MAX_TOKENS
                    ),

                    timeout=(
                        CELL55_REQUEST_TIMEOUT_SECONDS
                    ),

                    extra_body={
                        "enable_thinking": False,
                    },
                )
            )

            if response is None:
                raise RuntimeError(
                    "EMPTY_API_RESPONSE"
                )

            choices = getattr(
                response,
                "choices",
                None,
            )

            if (
                not isinstance(
                    choices,
                    (list, tuple),
                )
                or len(choices) == 0
            ):
                raise RuntimeError(
                    "API_RESPONSE_CHOICES_MISSING_OR_EMPTY"
                )

            choice = choices[0]

            message = getattr(
                choice,
                "message",
                None,
            )

            if message is None:
                raise RuntimeError(
                    "API_RESPONSE_MESSAGE_MISSING"
                )

            raw_content = getattr(
                message,
                "content",
                None,
            )

            finish_reason = getattr(
                choice,
                "finish_reason",
                None,
            )

            usage = getattr(
                response,
                "usage",
                None,
            )

            input_tokens = int(
                getattr(
                    usage,
                    "prompt_tokens",
                    0,
                )
                or 0
            )

            output_tokens = int(
                getattr(
                    usage,
                    "completion_tokens",
                    0,
                )
                or 0
            )

            total_tokens = int(
                getattr(
                    usage,
                    "total_tokens",
                    0,
                )
                or (
                    input_tokens
                    + output_tokens
                )
            )

            clean_content = (
                cell55_clean_json(
                    raw_content
                )
            )

            parsed_json = None
            parse_error = None

            try:

                parsed_json = json.loads(
                    clean_content
                )

            except Exception as exc:

                parse_error = (
                    f"{type(exc).__name__}: {exc}"
                )

            if parsed_json is None:

                structural_errors = [
                    "JSON_PARSE_FAILED"
                ]

            else:

                structural_errors = (
                    cell41_validate_generation_output_v4(
                        parsed_json
                    )
                )

            structural_status = (
                "PASS"
                if not structural_errors
                else "FAIL"
            )

            return {
                "candidate_id": (
                    CELL55_CANDIDATE_ID
                ),

                "request_fingerprint": (
                    CELL55_REQUEST_FINGERPRINT
                ),

                "execution_status": (
                    "COMPLETED"
                ),

                "api_attempts": (
                    attempt_number
                ),

                "attempt_history": (
                    attempt_history
                ),

                "finish_reason": (
                    finish_reason
                ),

                "usage": {
                    "input_tokens": (
                        input_tokens
                    ),

                    "output_tokens": (
                        output_tokens
                    ),

                    "total_tokens": (
                        total_tokens
                    ),
                },

                "raw_content": (
                    raw_content
                ),

                "parsed_json": (
                    parsed_json
                ),

                "parse_error": (
                    parse_error
                ),

                "structural_errors": (
                    structural_errors
                ),

                "structural_status": (
                    structural_status
                ),
            }

        except Exception as exc:

            attempt_history.append({
                "attempt": attempt_number,
                "status": "API_ERROR",
                "error": (
                    f"{type(exc).__name__}: {exc}"
                ),
            })

            if (
                attempt_number
                < CELL55_MAX_ATTEMPTS
            ):

                time.sleep(
                    CELL55_RETRY_SECONDS
                    * attempt_number
                )

                continue

            return {
                "candidate_id": (
                    CELL55_CANDIDATE_ID
                ),

                "request_fingerprint": (
                    CELL55_REQUEST_FINGERPRINT
                ),

                "execution_status": (
                    "API_FAILED"
                ),

                "api_attempts": (
                    attempt_number
                ),

                "attempt_history": (
                    attempt_history
                ),

                "finish_reason": None,

                "usage": {
                    "input_tokens": 0,
                    "output_tokens": 0,
                    "total_tokens": 0,
                },

                "raw_content": None,
                "parsed_json": None,
                "parse_error": None,

                "structural_errors": [
                    "API_FAILED"
                ],

                "structural_status": (
                    "FAIL"
                ),
            }


# =========================================================
# PREFLIGHT
# =========================================================

(
    cell55_existing_checkpoint,
    cell55_checkpoint_status_before,
    cell55_checkpoint_errors_before,
) = cell55_load_checkpoint()


assert (
    cell55_checkpoint_status_before
    != "INVALID"
), (
    "Invalid Cell 55 checkpoint detected; "
    "refusing automatic paid replacement."
)

assert (
    cell55_checkpoint_status_before
    != "UNREADABLE"
), (
    "Unreadable Cell 55 checkpoint detected; "
    "refusing automatic paid replacement."
)


cell55_api_candidates_this_execution = 0


print(
    "=== GOLDEN V2 FINAL SINGLE-CANDIDATE "
    "QA REPAIR PREFLIGHT ==="
)

print()

print(
    "Full population:",
    3813,
)

print(
    "Semantic PASS preserved:",
    3812,
)

print(
    "Repair population:",
    1,
)

print(
    "Candidate:",
    CELL55_CANDIDATE_ID,
)

print(
    "Existing exact checkpoint:",
    cell55_checkpoint_status_before,
)

print(
    "Model:",
    semantic_review_model,
)

print(
    "Max tokens:",
    CELL55_MAX_TOKENS,
)

print(
    "API allowed if checkpoint missing:",
    CELL55_ALLOW_API_FOR_MISSING,
)

print(
    "Previously accepted 3812 QA sent to API:",
    0,
)

print()


# =========================================================
# RUN OR RECONSTRUCT
# =========================================================

if (
    cell55_checkpoint_status_before
    == "VALID"
):

    cell55_final_checkpoint = (
        cell55_existing_checkpoint
    )

else:

    assert (
        CELL55_ALLOW_API_FOR_MISSING
    ), (
        "Cell 55 checkpoint missing "
        "and API execution disabled."
    )

    cell55_api_candidates_this_execution = 1

    cell55_result = (
        cell55_execute_api()
    )

    cell55_final_checkpoint = (
        cell55_build_checkpoint(
            cell55_result
        )
    )

    cell55_atomic_write_json(
        cell55_checkpoint_path,
        cell55_final_checkpoint,
    )


# =========================================================
# VERIFY CHECKPOINT
# =========================================================

(
    cell55_verified_checkpoint,
    cell55_verified_status,
    cell55_verified_errors,
) = cell55_load_checkpoint()


assert (
    cell55_verified_status
    == "VALID"
), (
    "Cell 55 checkpoint verification failed: "
    f"{cell55_verified_errors}"
)


cell55_final_repair_result = (
    cell55_verified_checkpoint[
        "result"
    ]
)


# =========================================================
# HEALTH
# =========================================================

cell55_checks = {
    "single_candidate_only": (
        cell55_final_repair_result[
            "candidate_id"
        ]
        == CELL55_CANDIDATE_ID
    ),

    "checkpoint_valid": (
        cell55_verified_status
        == "VALID"
    ),

    "execution_completed": (
        cell55_final_repair_result.get(
            "execution_status"
        )
        == "COMPLETED"
    ),

    "structural_pass": (
        cell55_final_repair_result.get(
            "structural_status"
        )
        == "PASS"
    ),

    "finish_stop": (
        cell55_final_repair_result.get(
            "finish_reason"
        )
        == "stop"
    ),

    "preserved_3812_not_sent_to_api": True,
}


cell55_status = (
    "PASS"
    if all(
        cell55_checks.values()
    )
    else "FAIL"
)


cell55_next_readiness = (
    "READY_FOR_FINAL_SINGLE_CANDIDATE_SEMANTIC_REVALIDATION"
    if cell55_status == "PASS"
    else
    "READY_FOR_CELL55_REPAIR_ANALYSIS"
)


# =========================================================
# OUTPUT
# =========================================================

print(
    "=== GOLDEN V2 FINAL SINGLE-CANDIDATE "
    "QA REPAIR RESULT ==="
)

print()

print(
    "API candidates this execution:",
    cell55_api_candidates_this_execution,
)

print(
    "API attempts:",
    cell55_final_repair_result.get(
        "api_attempts"
    ),
)

print(
    "Execution status:",
    cell55_final_repair_result.get(
        "execution_status"
    ),
)

print(
    "Finish reason:",
    cell55_final_repair_result.get(
        "finish_reason"
    ),
)

print(
    "Structural status:",
    cell55_final_repair_result.get(
        "structural_status"
    ),
)

print()

print(
    "Final Question:",
    (
        cell55_final_repair_result
        .get(
            "parsed_json",
            {}
        )
        .get(
            "question"
        )
    ),
)

print(
    "Final Expected Answer:",
    (
        cell55_final_repair_result
        .get(
            "parsed_json",
            {}
        )
        .get(
            "expected_answer"
        )
    ),
)

print(
    "Final Question Intent:",
    (
        cell55_final_repair_result
        .get(
            "parsed_json",
            {}
        )
        .get(
            "question_intent"
        )
    ),
)

print()

print("Checks:")

for (
    check_name,
    check_value,
) in cell55_checks.items():

    print(
        " -",
        check_name,
        ":",
        (
            "PASS"
            if check_value
            else "FAIL"
        ),
    )

print()

print(
    "CELL 55 STATUS:",
    cell55_status,
)

print(
    "Next readiness:",
    cell55_next_readiness,
)


assert cell55_status == "PASS", (
    "Cell 55 final QA repair did not reach "
    "a structurally valid completed result."
)
````

````python
# Code Cell 56 — Final Single-Candidate Semantic Revalidation

from pathlib import Path
import hashlib
import json
import os
import re
import tempfile
import time


# =========================================================
# CONFIG
# =========================================================

CELL56_RUNNER_VERSION = (
    "golden_v2_final_single_candidate_semantic_revalidation_v1"
)

CELL56_VALIDATION_PROTOCOL_VERSION = (
    CELL53_VALIDATION_PROTOCOL_VERSION
)

CELL56_OUTPUT_SCHEMA_VERSION = (
    CELL53_OUTPUT_SCHEMA_VERSION
)

CELL56_MAX_TOKENS = 3072
CELL56_TEMPERATURE = 0.0
CELL56_TOP_P = 1.0

CELL56_MAX_ATTEMPTS = 2
CELL56_RETRY_SECONDS = 2.0
CELL56_REQUEST_TIMEOUT_SECONDS = 180.0

# Before final Restart -> Run All this MUST become False.
CELL56_ALLOW_API_FOR_MISSING = False


# =========================================================
# REQUIRED STATE
# =========================================================

cell56_required_names = [
    "project_root",
    "semantic_review_client",
    "semantic_review_model",
    "cell25_candidate_skeletons",
    "cell53_final_semantic_results",
    "cell55_final_repair_result",
    "CELL37_SYSTEM_PROMPT",
    "CELL37_BOOLEAN_FIELDS",
    "cell37_validate_output",
]

cell56_missing_state = [
    name
    for name in cell56_required_names
    if name not in globals()
]

assert not cell56_missing_state, (
    "Missing required Cell 56 state: "
    + ", ".join(cell56_missing_state)
)


# =========================================================
# EXACT SINGLE TARGET
# =========================================================

CELL56_CANDIDATE_ID = (
    "golden_v2_candidate_f4b2f9e2cb3874670125b227"
)

assert len(
    cell53_final_semantic_results
) == 3813

assert (
    cell55_final_repair_result.get(
        "candidate_id"
    )
    == CELL56_CANDIDATE_ID
)

assert (
    cell55_final_repair_result.get(
        "structural_status"
    )
    == "PASS"
)

assert (
    cell55_final_repair_result.get(
        "execution_status"
    )
    == "COMPLETED"
)


# =========================================================
# LOOKUPS
# =========================================================

cell56_skeleton_by_candidate_id = {
    record["candidate_id"]: record
    for record in cell25_candidate_skeletons
}

cell56_previous_semantic_by_candidate_id = {
    result["candidate_id"]: result
    for result in cell53_final_semantic_results
}

assert (
    CELL56_CANDIDATE_ID
    in cell56_skeleton_by_candidate_id
)

assert (
    CELL56_CANDIDATE_ID
    in cell56_previous_semantic_by_candidate_id
)


cell56_skeleton = (
    cell56_skeleton_by_candidate_id[
        CELL56_CANDIDATE_ID
    ]
)

cell56_final_qa = (
    cell55_final_repair_result[
        "parsed_json"
    ]
)

assert isinstance(
    cell56_final_qa,
    dict,
)


# =========================================================
# PRESERVE 3812
# =========================================================

cell56_preserved_semantic_results = [
    result
    for result
    in cell53_final_semantic_results
    if (
        result["candidate_id"]
        != CELL56_CANDIDATE_ID
    )
]

assert len(
    cell56_preserved_semantic_results
) == 3812

assert all(
    result.get("structural_status") == "PASS"
    and
    result.get("semantic_verdict") == "PASS"
    for result
    in cell56_preserved_semantic_results
)


# =========================================================
# HASH UTILITIES
# =========================================================

def cell56_canonical_json(value):

    return json.dumps(
        value,
        ensure_ascii=False,
        sort_keys=True,
        separators=(",", ":"),
    )


def cell56_sha256_text(text):

    return hashlib.sha256(
        text.encode("utf-8")
    ).hexdigest()


# =========================================================
# VALIDATION PACKET
# =========================================================

cell56_evidence_ids = list(
    cell56_skeleton["evidence_ids"]
)

cell56_evidence_quotes = list(
    cell56_skeleton["evidence_quotes"]
)

assert len(
    cell56_evidence_ids
) == len(
    cell56_evidence_quotes
)


cell56_evidence = [
    {
        "index": index,
        "evidence_id": evidence_id,
        "quote": evidence_quote,
    }
    for index, (
        evidence_id,
        evidence_quote,
    )
    in enumerate(
        zip(
            cell56_evidence_ids,
            cell56_evidence_quotes,
        ),
        start=1,
    )
]


cell56_validation_packet = {
    "candidate_id": (
        CELL56_CANDIDATE_ID
    ),

    "claim": {
        "claim_id": (
            cell56_skeleton[
                "claim_id"
            ]
        ),

        "claim_text_fa": (
            cell56_skeleton[
                "claim_text_fa"
            ]
        ),

        "claim_type": (
            cell56_skeleton.get(
                "claim_type"
            )
        ),

        "claim_types": (
            cell56_skeleton.get(
                "claim_types",
                [],
            )
        ),

        "authority_class": (
            cell56_skeleton.get(
                "authority_class"
            )
        ),
    },

    "evidence": (
        cell56_evidence
    ),

    "generated_qa": {
        "question": (
            cell56_final_qa[
                "question"
            ]
        ),

        "expected_answer": (
            cell56_final_qa[
                "expected_answer"
            ]
        ),

        "question_intent": (
            cell56_final_qa[
                "question_intent"
            ]
        ),
    },
}


# =========================================================
# VALIDATOR PROMPT
# =========================================================

CELL56_USER_PROMPT = (
    "Validate ONLY the generated Question and Expected Answer "
    "against the supplied prevalidated Claim and Evidence.\n\n"

    "The Claim and Evidence have already passed upstream semantic "
    "validation. Do NOT decide whether the Claim itself should be "
    "accepted or rejected.\n\n"

    "Evaluate exactly these QA dimensions:\n"
    "1. question_targets_claim\n"
    "2. answer_recovers_claim\n"
    "3. scope_fidelity\n"
    "4. modality_quantifier_fidelity\n"
    "5. standalone_question\n"
    "6. no_neighbor_fact_drift\n"
    "7. no_external_knowledge\n"
    "8. answer_question_alignment\n\n"

    "Important fidelity rules:\n"
    "- The Question must ask for information actually asserted "
    "by the Claim.\n"
    "- The Question must directly identify enough of the Claim's "
    "subject to be standalone rather than context-dependent.\n"
    "- The Expected Answer must directly answer the Question and "
    "faithfully recover the relevant meaning of the Claim.\n"
    "- Preserve scope, modality, attribution, conditions, "
    "negation and quantifiers.\n"
    "- Do not allow WHY/HOW/cause/mechanism unless asserted.\n"
    "- Do not allow neighboring facts or external knowledge.\n\n"

    "Return only the JSON object required by the validation "
    "system prompt.\n\n"

    "VALIDATION PACKET:\n"
    + json.dumps(
        cell56_validation_packet,
        ensure_ascii=False,
        sort_keys=True,
        indent=2,
    )
)


# =========================================================
# REQUEST CONTRACT
# =========================================================

cell56_request_contract = {
    "runner_version": (
        CELL56_RUNNER_VERSION
    ),

    "validation_protocol_version": (
        CELL56_VALIDATION_PROTOCOL_VERSION
    ),

    "output_schema_version": (
        CELL56_OUTPUT_SCHEMA_VERSION
    ),

    "candidate_id": (
        CELL56_CANDIDATE_ID
    ),

    "qa_sha256": (
        cell56_sha256_text(
            cell56_canonical_json(
                cell56_final_qa
            )
        )
    ),

    "validation_packet_sha256": (
        cell56_sha256_text(
            cell56_canonical_json(
                cell56_validation_packet
            )
        )
    ),

    "model": (
        semantic_review_model
    ),

    "system_prompt": (
        CELL37_SYSTEM_PROMPT
    ),

    "user_prompt": (
        CELL56_USER_PROMPT
    ),

    "temperature": (
        CELL56_TEMPERATURE
    ),

    "top_p": (
        CELL56_TOP_P
    ),

    "max_tokens": (
        CELL56_MAX_TOKENS
    ),

    "enable_thinking": False,
}


CELL56_REQUEST_FINGERPRINT = (
    cell56_sha256_text(
        cell56_canonical_json(
            cell56_request_contract
        )
    )
)


# =========================================================
# CHECKPOINT
# =========================================================

cell56_checkpoint_dir = (
    Path(project_root)
    / "artifacts"
    / "phase2"
    / "golden_dataset_v2"
    / "work"
    / "final_single_candidate_semantic_revalidation_v1"
)

cell56_checkpoint_dir.mkdir(
    parents=True,
    exist_ok=True,
)

cell56_checkpoint_path = (
    cell56_checkpoint_dir
    / f"{CELL56_CANDIDATE_ID}.json"
)


def cell56_atomic_write_json(
    path,
    payload,
):

    path = Path(path)

    content = (
        json.dumps(
            payload,
            ensure_ascii=False,
            sort_keys=True,
            indent=2,
        )
        + "\n"
    )

    fd, temp_name = tempfile.mkstemp(
        prefix=path.name + ".",
        suffix=".tmp",
        dir=str(path.parent),
    )

    try:

        with os.fdopen(
            fd,
            "w",
            encoding="utf-8",
        ) as handle:

            handle.write(content)
            handle.flush()
            os.fsync(handle.fileno())

        os.replace(
            temp_name,
            path,
        )

    except Exception:

        try:
            os.unlink(temp_name)
        except FileNotFoundError:
            pass

        raise


def cell56_build_checkpoint(result):

    checkpoint = {
        "checkpoint_version": (
            CELL56_RUNNER_VERSION
        ),

        "candidate_id": (
            CELL56_CANDIDATE_ID
        ),

        "request_fingerprint": (
            CELL56_REQUEST_FINGERPRINT
        ),

        "qa_sha256": (
            cell56_request_contract[
                "qa_sha256"
            ]
        ),

        "validation_packet_sha256": (
            cell56_request_contract[
                "validation_packet_sha256"
            ]
        ),

        "validation_protocol_version": (
            CELL56_VALIDATION_PROTOCOL_VERSION
        ),

        "output_schema_version": (
            CELL56_OUTPUT_SCHEMA_VERSION
        ),

        "model": (
            semantic_review_model
        ),

        "temperature": (
            CELL56_TEMPERATURE
        ),

        "top_p": (
            CELL56_TOP_P
        ),

        "max_tokens": (
            CELL56_MAX_TOKENS
        ),

        "result": result,
    }

    checkpoint[
        "checkpoint_payload_sha256"
    ] = cell56_sha256_text(
        cell56_canonical_json(
            checkpoint
        )
    )

    return checkpoint


def cell56_validate_checkpoint(
    checkpoint,
):

    errors = []

    if not isinstance(
        checkpoint,
        dict,
    ):

        return [
            "CHECKPOINT_NOT_OBJECT"
        ]

    stored_sha = checkpoint.get(
        "checkpoint_payload_sha256"
    )

    payload_without_sha = {
        key: value
        for key, value
        in checkpoint.items()
        if key
        != "checkpoint_payload_sha256"
    }

    recalculated_sha = (
        cell56_sha256_text(
            cell56_canonical_json(
                payload_without_sha
            )
        )
    )

    if stored_sha != recalculated_sha:

        errors.append(
            "CHECKPOINT_HASH_MISMATCH"
        )

    expected = {
        "checkpoint_version": (
            CELL56_RUNNER_VERSION
        ),

        "candidate_id": (
            CELL56_CANDIDATE_ID
        ),

        "request_fingerprint": (
            CELL56_REQUEST_FINGERPRINT
        ),

        "qa_sha256": (
            cell56_request_contract[
                "qa_sha256"
            ]
        ),

        "validation_packet_sha256": (
            cell56_request_contract[
                "validation_packet_sha256"
            ]
        ),

        "validation_protocol_version": (
            CELL56_VALIDATION_PROTOCOL_VERSION
        ),

        "output_schema_version": (
            CELL56_OUTPUT_SCHEMA_VERSION
        ),

        "model": (
            semantic_review_model
        ),

        "temperature": (
            CELL56_TEMPERATURE
        ),

        "top_p": (
            CELL56_TOP_P
        ),

        "max_tokens": (
            CELL56_MAX_TOKENS
        ),
    }

    for (
        field_name,
        expected_value,
    ) in expected.items():

        if (
            checkpoint.get(
                field_name
            )
            != expected_value
        ):

            errors.append(
                f"{field_name.upper()}_MISMATCH"
            )

    result = checkpoint.get(
        "result"
    )

    if not isinstance(
        result,
        dict,
    ):

        errors.append(
            "RESULT_INVALID"
        )

    elif (
        result.get(
            "candidate_id"
        )
        != CELL56_CANDIDATE_ID
    ):

        errors.append(
            "RESULT_CANDIDATE_MISMATCH"
        )

    return sorted(
        set(errors)
    )


def cell56_load_checkpoint():

    if not cell56_checkpoint_path.exists():

        return (
            None,
            "MISSING",
            [],
        )

    try:

        with cell56_checkpoint_path.open(
            "r",
            encoding="utf-8",
        ) as handle:

            checkpoint = json.load(
                handle
            )

    except Exception as exc:

        return (
            None,
            "UNREADABLE",
            [
                f"{type(exc).__name__}: {exc}"
            ],
        )

    errors = (
        cell56_validate_checkpoint(
            checkpoint
        )
    )

    if errors:

        return (
            checkpoint,
            "INVALID",
            errors,
        )

    return (
        checkpoint,
        "VALID",
        [],
    )


# =========================================================
# RESPONSE UTILITIES
# =========================================================

def cell56_clean_json(content):

    if not isinstance(
        content,
        str,
    ):
        return ""

    clean = content.strip()

    if clean.startswith("```"):

        clean = re.sub(
            r"^```(?:json)?\s*",
            "",
            clean,
            flags=re.IGNORECASE,
        )

        clean = re.sub(
            r"\s*```$",
            "",
            clean,
        )

    return clean.strip()


# =========================================================
# API EXECUTION
# =========================================================

def cell56_execute_api():

    attempt_history = []

    for attempt_number in range(
        1,
        CELL56_MAX_ATTEMPTS + 1,
    ):

        try:

            response = (
                semantic_review_client
                .chat
                .completions
                .create(
                    model=semantic_review_model,

                    messages=[
                        {
                            "role": "system",
                            "content": (
                                CELL37_SYSTEM_PROMPT
                            ),
                        },
                        {
                            "role": "user",
                            "content": (
                                CELL56_USER_PROMPT
                            ),
                        },
                    ],

                    temperature=(
                        CELL56_TEMPERATURE
                    ),

                    top_p=(
                        CELL56_TOP_P
                    ),

                    max_tokens=(
                        CELL56_MAX_TOKENS
                    ),

                    timeout=(
                        CELL56_REQUEST_TIMEOUT_SECONDS
                    ),

                    extra_body={
                        "enable_thinking": False,
                    },
                )
            )

            if response is None:

                raise RuntimeError(
                    "EMPTY_API_RESPONSE"
                )

            choices = getattr(
                response,
                "choices",
                None,
            )

            if (
                not isinstance(
                    choices,
                    (list, tuple),
                )
                or len(choices) == 0
            ):

                raise RuntimeError(
                    "API_RESPONSE_CHOICES_MISSING_OR_EMPTY"
                )

            choice = choices[0]

            message = getattr(
                choice,
                "message",
                None,
            )

            if message is None:

                raise RuntimeError(
                    "API_RESPONSE_MESSAGE_MISSING"
                )

            raw_content = getattr(
                message,
                "content",
                None,
            )

            finish_reason = getattr(
                choice,
                "finish_reason",
                None,
            )

            usage = getattr(
                response,
                "usage",
                None,
            )

            input_tokens = int(
                getattr(
                    usage,
                    "prompt_tokens",
                    0,
                )
                or 0
            )

            output_tokens = int(
                getattr(
                    usage,
                    "completion_tokens",
                    0,
                )
                or 0
            )

            total_tokens = int(
                getattr(
                    usage,
                    "total_tokens",
                    0,
                )
                or (
                    input_tokens
                    + output_tokens
                )
            )

            clean_content = (
                cell56_clean_json(
                    raw_content
                )
            )

            parsed_json = None
            parse_error = None

            try:

                parsed_json = json.loads(
                    clean_content
                )

            except Exception as exc:

                parse_error = (
                    f"{type(exc).__name__}: {exc}"
                )

            if parsed_json is None:

                structural_errors = [
                    "JSON_PARSE_FAILED"
                ]

            else:

                structural_errors = (
                    cell37_validate_output(
                        parsed_json
                    )
                )

            structural_status = (
                "PASS"
                if not structural_errors
                else "FAIL"
            )

            semantic_verdict = None

            if (
                structural_status
                == "PASS"
            ):

                semantic_verdict = (
                    parsed_json.get(
                        "verdict"
                    )
                )

            return {
                "candidate_id": (
                    CELL56_CANDIDATE_ID
                ),

                "request_fingerprint": (
                    CELL56_REQUEST_FINGERPRINT
                ),

                "execution_status": (
                    "COMPLETED"
                ),

                "api_attempts": (
                    attempt_number
                ),

                "attempt_history": (
                    attempt_history
                ),

                "finish_reason": (
                    finish_reason
                ),

                "usage": {
                    "input_tokens": (
                        input_tokens
                    ),

                    "output_tokens": (
                        output_tokens
                    ),

                    "total_tokens": (
                        total_tokens
                    ),
                },

                "raw_content": (
                    raw_content
                ),

                "parsed_json": (
                    parsed_json
                ),

                "parse_error": (
                    parse_error
                ),

                "structural_errors": (
                    structural_errors
                ),

                "structural_status": (
                    structural_status
                ),

                "semantic_verdict": (
                    semantic_verdict
                ),

                "validation": (
                    parsed_json
                    if structural_status
                    == "PASS"
                    else None
                ),
            }

        except Exception as exc:

            attempt_history.append({
                "attempt": (
                    attempt_number
                ),

                "status": (
                    "API_ERROR"
                ),

                "error": (
                    f"{type(exc).__name__}: {exc}"
                ),
            })

            if (
                attempt_number
                < CELL56_MAX_ATTEMPTS
            ):

                time.sleep(
                    CELL56_RETRY_SECONDS
                    * attempt_number
                )

                continue

            return {
                "candidate_id": (
                    CELL56_CANDIDATE_ID
                ),

                "request_fingerprint": (
                    CELL56_REQUEST_FINGERPRINT
                ),

                "execution_status": (
                    "API_FAILED"
                ),

                "api_attempts": (
                    attempt_number
                ),

                "attempt_history": (
                    attempt_history
                ),

                "finish_reason": None,

                "usage": {
                    "input_tokens": 0,
                    "output_tokens": 0,
                    "total_tokens": 0,
                },

                "raw_content": None,
                "parsed_json": None,
                "parse_error": None,

                "structural_errors": [
                    "API_FAILED"
                ],

                "structural_status": (
                    "FAIL"
                ),

                "semantic_verdict": None,
                "validation": None,
            }


# =========================================================
# PREFLIGHT
# =========================================================

(
    cell56_existing_checkpoint,
    cell56_checkpoint_status_before,
    cell56_checkpoint_errors_before,
) = cell56_load_checkpoint()


assert (
    cell56_checkpoint_status_before
    not in {
        "INVALID",
        "UNREADABLE",
    }
), (
    "Invalid/unreadable Cell 56 checkpoint detected. "
    "Refusing automatic paid replacement."
)


print(
    "=== GOLDEN V2 FINAL SINGLE-CANDIDATE "
    "SEMANTIC REVALIDATION PREFLIGHT ==="
)

print()

print(
    "Full population:",
    3813,
)

print(
    "Semantic PASS preserved:",
    3812,
)

print(
    "Revalidation population:",
    1,
)

print(
    "Candidate:",
    CELL56_CANDIDATE_ID,
)

print(
    "Existing exact checkpoint:",
    cell56_checkpoint_status_before,
)

print(
    "Model:",
    semantic_review_model,
)

print(
    "API allowed if checkpoint missing:",
    CELL56_ALLOW_API_FOR_MISSING,
)

print(
    "Previously accepted 3812 QA sent to API:",
    0,
)

print()


# =========================================================
# RUN OR RECONSTRUCT
# =========================================================

cell56_api_candidates_this_execution = 0


if (
    cell56_checkpoint_status_before
    == "VALID"
):

    cell56_final_checkpoint = (
        cell56_existing_checkpoint
    )

else:

    assert (
        CELL56_ALLOW_API_FOR_MISSING
    ), (
        "Cell 56 checkpoint missing "
        "and API execution disabled."
    )

    cell56_api_candidates_this_execution = 1

    cell56_result = (
        cell56_execute_api()
    )

    cell56_final_checkpoint = (
        cell56_build_checkpoint(
            cell56_result
        )
    )

    cell56_atomic_write_json(
        cell56_checkpoint_path,
        cell56_final_checkpoint,
    )


# =========================================================
# VERIFY
# =========================================================

(
    cell56_verified_checkpoint,
    cell56_verified_status,
    cell56_verified_errors,
) = cell56_load_checkpoint()


assert (
    cell56_verified_status
    == "VALID"
), (
    "Cell 56 checkpoint verification failed: "
    f"{cell56_verified_errors}"
)


cell56_final_validation_result = (
    cell56_verified_checkpoint[
        "result"
    ]
)


# =========================================================
# MERGE FINAL 3813
# =========================================================

cell56_final_semantic_results = []


for result in (
    cell53_final_semantic_results
):

    candidate_id = (
        result["candidate_id"]
    )

    if (
        candidate_id
        == CELL56_CANDIDATE_ID
    ):

        cell56_final_semantic_results.append(
            cell56_final_validation_result
        )

    else:

        cell56_final_semantic_results.append(
            result
        )


assert len(
    cell56_final_semantic_results
) == 3813


# =========================================================
# FINAL COUNTS
# =========================================================

cell56_final_structural_pass_count = sum(
    1
    for result
    in cell56_final_semantic_results
    if result.get(
        "structural_status"
    )
    == "PASS"
)

cell56_final_structural_fail_count = (
    3813
    - cell56_final_structural_pass_count
)

cell56_final_api_failure_count = sum(
    1
    for result
    in cell56_final_semantic_results
    if result.get(
        "execution_status"
    )
    == "API_FAILED"
)

cell56_final_semantic_pass_count = sum(
    1
    for result
    in cell56_final_semantic_results
    if (
        result.get(
            "structural_status"
        )
        == "PASS"
        and
        result.get(
            "semantic_verdict"
        )
        == "PASS"
    )
)

cell56_final_semantic_fail_count = sum(
    1
    for result
    in cell56_final_semantic_results
    if (
        result.get(
            "structural_status"
        )
        == "PASS"
        and
        result.get(
            "semantic_verdict"
        )
        == "FAIL"
    )
)


# =========================================================
# HEALTH
# =========================================================

cell56_checks = {
    "full_population_3813": (
        len(
            cell56_final_semantic_results
        )
        == 3813
    ),

    "single_candidate_revalidated": (
        cell56_final_validation_result[
            "candidate_id"
        ]
        == CELL56_CANDIDATE_ID
    ),

    "execution_completed": (
        cell56_final_validation_result.get(
            "execution_status"
        )
        == "COMPLETED"
    ),

    "final_candidate_structural_pass": (
        cell56_final_validation_result.get(
            "structural_status"
        )
        == "PASS"
    ),

    "preserved_3812_not_sent_to_api": True,

    "checkpoint_valid": (
        cell56_verified_status
        == "VALID"
    ),
}


cell56_runner_status = (
    "PASS"
    if all(
        cell56_checks.values()
    )
    else "FAIL"
)


if cell56_runner_status != "PASS":

    cell56_next_readiness = (
        "BLOCKED_CELL56_RUNNER_REPAIR_REQUIRED"
    )

elif (
    cell56_final_validation_result.get(
        "execution_status"
    )
    == "API_FAILED"
):

    cell56_next_readiness = (
        "READY_FOR_CELL56_TECHNICAL_RECOVERY"
    )

elif (
    cell56_final_validation_result.get(
        "structural_status"
    )
    != "PASS"
):

    cell56_next_readiness = (
        "READY_FOR_CELL56_STRUCTURAL_RECOVERY"
    )

elif (
    cell56_final_validation_result.get(
        "semantic_verdict"
    )
    == "FAIL"
):

    cell56_next_readiness = (
        "READY_FOR_FINAL_FAILURE_DISPOSITION"
    )

elif (
    cell56_final_semantic_pass_count
    == 3813
    and
    cell56_final_semantic_fail_count
    == 0
):

    cell56_next_readiness = (
        "READY_FOR_GOLDEN_V2_FINAL_RECORD_CONSTRUCTION"
    )

else:

    cell56_next_readiness = (
        "READY_FOR_CELL56_RESULT_ANALYSIS"
    )


# =========================================================
# OUTPUT
# =========================================================

print(
    "=== GOLDEN V2 FINAL SINGLE-CANDIDATE "
    "SEMANTIC REVALIDATION RESULT ==="
)

print()

print(
    "API candidates this execution:",
    cell56_api_candidates_this_execution,
)

print(
    "API attempts:",
    cell56_final_validation_result.get(
        "api_attempts"
    ),
)

print(
    "Execution status:",
    cell56_final_validation_result.get(
        "execution_status"
    ),
)

print(
    "Finish reason:",
    cell56_final_validation_result.get(
        "finish_reason"
    ),
)

print(
    "Structural status:",
    cell56_final_validation_result.get(
        "structural_status"
    ),
)

print(
    "Semantic verdict:",
    cell56_final_validation_result.get(
        "semantic_verdict"
    ),
)

print()

print(
    "Validation:"
)

print(
    json.dumps(
        cell56_final_validation_result.get(
            "validation"
        ),
        ensure_ascii=False,
        indent=2,
        sort_keys=True,
    )
)

print()

print(
    "=== FINAL 3813 STATE ==="
)

print(
    "Structural PASS:",
    cell56_final_structural_pass_count,
)

print(
    "Structural FAIL:",
    cell56_final_structural_fail_count,
)

print(
    "API failures:",
    cell56_final_api_failure_count,
)

print(
    "Semantic PASS:",
    cell56_final_semantic_pass_count,
)

print(
    "Semantic FAIL:",
    cell56_final_semantic_fail_count,
)

print()

print("Runner checks:")

for (
    check_name,
    check_value,
) in cell56_checks.items():

    print(
        " -",
        check_name,
        ":",
        (
            "PASS"
            if check_value
            else "FAIL"
        ),
    )

print()

print(
    "Runner health:",
    cell56_runner_status,
)

print(
    "Next readiness:",
    cell56_next_readiness,
)

print(
    "CELL 56 CHECKPOINTED RUNNER:",
    cell56_runner_status,
)


assert cell56_runner_status == "PASS", (
    "Cell 56 runner health failed."
)
````

<h2>مرحله ۵۷ — ساخت قطعی رکوردهای نهایی Golden Dataset v2</h2>
<p>
مرحله اعتبارسنجی Question/Answer با موفقیت کامل بسته شد: هر ۳۸۱۳ Candidate دارای Structural PASS و Semantic PASS هستند و هیچ Semantic Failure یا API Failure باقی نمانده است. از این مرحله به بعد برای ساخت Dataset نهایی نیازی به فراخوانی مدل وجود ندارد.
</p>
<p>
در این مرحله، Skeletonهای معتبر مرحله ۲۵ با Question/Answer نهایی هر Candidate ادغام می‌شوند. برای Candidateهایی که Repair نشده‌اند، QA معتبر حاصل از Full Generation استفاده می‌شود؛ برای ۳۶ Candidate اصلاح‌شده و تأییدشده، خروجی Repair نهایی مرحله ۵۲ استفاده می‌شود؛ و برای آخرین Candidate، خروجی Repair نهایی مرحله ۵۵ جایگزین می‌شود. Judgment معنایی نهایی نیز از وضعیت نهایی مرحله ۵۶ گرفته می‌شود.
</p>
<p>
ساخت رکوردها کاملاً Deterministic، بدون API و بدون File Write انجام می‌شود. هویت Candidate، Claim، Evidence و Source Lineage موجود در Skeleton حفظ می‌شوند و فقط Evaluation Content و وضعیت Generation/Validation با خروجی نهایی معتبر تکمیل می‌شوند. این مرحله همچنین یکتایی ۳۸۱۳ Candidate، وجود Question و Expected Answer معتبر و Semantic PASS بودن همه رکوردها را کنترل می‌کند. Freeze و نوشتن فایل Golden Dataset v2 در مرحله بعد و پس از عبور این Gate انجام خواهد شد.
</p>
</div>

````python
# Code Cell 57 — Deterministic Final Golden Dataset v2 Record Construction

from collections import Counter
from copy import deepcopy
import hashlib
import json


# =========================================================
# CONFIG
# =========================================================

CELL57_VERSION = (
    "golden_dataset_v2_final_record_construction_v1"
)

CELL57_DATASET_VERSION = (
    "golden_dataset_v2_registry_driven_v2"
)


# =========================================================
# REQUIRED STATE
# =========================================================

cell57_required_names = [
    "cell25_candidate_skeletons",
    "cell47_final_generation_results",
    "cell50_repair_candidate_ids",
    "cell52_final_repair_results",
    "cell55_final_repair_result",
    "cell56_final_semantic_results",
]

cell57_missing_state = [
    name
    for name in cell57_required_names
    if name not in globals()
]

assert not cell57_missing_state, (
    "Missing required Cell 57 state: "
    + ", ".join(cell57_missing_state)
)


# =========================================================
# INPUT POPULATION CONTRACT
# =========================================================

assert len(
    cell25_candidate_skeletons
) == 3813

assert len(
    cell47_final_generation_results
) == 3813

assert len(
    cell50_repair_candidate_ids
) == 37

assert len(
    cell52_final_repair_results
) == 37

assert len(
    cell56_final_semantic_results
) == 3813


# =========================================================
# LOOKUPS
# =========================================================

cell57_generation_by_candidate_id = {
    result["candidate_id"]: result
    for result
    in cell47_final_generation_results
}

cell57_repair_by_candidate_id = {
    result["candidate_id"]: result
    for result
    in cell52_final_repair_results
}

cell57_semantic_by_candidate_id = {
    result["candidate_id"]: result
    for result
    in cell56_final_semantic_results
}


assert len(
    cell57_generation_by_candidate_id
) == 3813

assert len(
    cell57_repair_by_candidate_id
) == 37

assert len(
    cell57_semantic_by_candidate_id
) == 3813


cell57_skeleton_candidate_ids = [
    record["candidate_id"]
    for record
    in cell25_candidate_skeletons
]

assert len(
    set(
        cell57_skeleton_candidate_ids
    )
) == 3813


assert (
    set(
        cell57_generation_by_candidate_id
    )
    ==
    set(
        cell57_skeleton_candidate_ids
    )
)

assert (
    set(
        cell57_semantic_by_candidate_id
    )
    ==
    set(
        cell57_skeleton_candidate_ids
    )
)

assert (
    set(
        cell57_repair_by_candidate_id
    )
    ==
    set(
        cell50_repair_candidate_ids
    )
)


# =========================================================
# FINAL QA SELECTION
# =========================================================

CELL57_FINAL_SINGLE_REPAIR_ID = (
    "golden_v2_candidate_f4b2f9e2cb3874670125b227"
)

assert (
    cell55_final_repair_result[
        "candidate_id"
    ]
    == CELL57_FINAL_SINGLE_REPAIR_ID
)

assert (
    cell55_final_repair_result.get(
        "structural_status"
    )
    == "PASS"
)


def cell57_select_final_generation(
    candidate_id,
):

    if (
        candidate_id
        == CELL57_FINAL_SINGLE_REPAIR_ID
    ):

        return (
            cell55_final_repair_result,
            "FINAL_SINGLE_CANDIDATE_REPAIR"
        )

    if (
        candidate_id
        in cell57_repair_by_candidate_id
    ):

        repair_result = (
            cell57_repair_by_candidate_id[
                candidate_id
            ]
        )

        assert (
            repair_result.get(
                "structural_status"
            )
            == "PASS"
        )

        return (
            repair_result,
            "TARGETED_SEMANTIC_REPAIR"
        )

    generation_result = (
        cell57_generation_by_candidate_id[
            candidate_id
        ]
    )

    assert (
        generation_result.get(
            "structural_status"
        )
        == "PASS"
    )

    return (
        generation_result,
        "FULL_GENERATION_V4"
    )


# =========================================================
# HASH UTILITIES
# =========================================================

def cell57_canonical_json(value):

    return json.dumps(
        value,
        ensure_ascii=False,
        sort_keys=True,
        separators=(",", ":"),
    )


def cell57_sha256_json(value):

    return hashlib.sha256(
        cell57_canonical_json(
            value
        ).encode("utf-8")
    ).hexdigest()


# =========================================================
# FINAL RECORD CONSTRUCTION
# =========================================================

cell57_final_records = []
cell57_record_errors = []
cell57_generation_source_counter = Counter()


for skeleton in (
    cell25_candidate_skeletons
):

    candidate_id = (
        skeleton["candidate_id"]
    )

    (
        final_generation_result,
        generation_source,
    ) = cell57_select_final_generation(
        candidate_id
    )

    final_qa = (
        final_generation_result.get(
            "parsed_json"
        )
    )

    final_semantic_result = (
        cell57_semantic_by_candidate_id[
            candidate_id
        ]
    )

    errors = []

    if not isinstance(
        final_qa,
        dict,
    ):

        errors.append(
            "FINAL_QA_NOT_OBJECT"
        )

        final_qa = {}

    question = final_qa.get(
        "question"
    )

    expected_answer = final_qa.get(
        "expected_answer"
    )

    question_intent = final_qa.get(
        "question_intent"
    )

    if (
        not isinstance(
            question,
            str,
        )
        or not question.strip()
    ):

        errors.append(
            "QUESTION_MISSING_OR_EMPTY"
        )

    if (
        not isinstance(
            expected_answer,
            str,
        )
        or not expected_answer.strip()
    ):

        errors.append(
            "EXPECTED_ANSWER_MISSING_OR_EMPTY"
        )

    if (
        not isinstance(
            question_intent,
            str,
        )
        or not question_intent.strip()
    ):

        errors.append(
            "QUESTION_INTENT_MISSING_OR_EMPTY"
        )

    if (
        final_generation_result.get(
            "structural_status"
        )
        != "PASS"
    ):

        errors.append(
            "FINAL_GENERATION_NOT_STRUCTURAL_PASS"
        )

    if (
        final_semantic_result.get(
            "structural_status"
        )
        != "PASS"
    ):

        errors.append(
            "FINAL_VALIDATION_NOT_STRUCTURAL_PASS"
        )

    if (
        final_semantic_result.get(
            "semantic_verdict"
        )
        != "PASS"
    ):

        errors.append(
            "FINAL_SEMANTIC_VERDICT_NOT_PASS"
        )

    if errors:

        cell57_record_errors.append({
            "candidate_id": (
                candidate_id
            ),
            "errors": errors,
        })

        continue

    final_record = deepcopy(
        skeleton
    )

    # ---------------------------------------------
    # Evaluation content
    # ---------------------------------------------

    final_record[
        "evaluation_content"
    ] = {
        "question": (
            question.strip()
        ),

        "expected_answer": (
            expected_answer.strip()
        ),

        "question_intent": (
            question_intent.strip()
        ),
    }

    # ---------------------------------------------
    # Generation provenance
    # ---------------------------------------------

    final_record[
        "generation"
    ] = {
        "status": "PASS",

        "source": (
            generation_source
        ),

        "protocol_version": (
            "golden_v2_question_answer_generation_v4"
        ),

        "structural_status": (
            final_generation_result.get(
                "structural_status"
            )
        ),

        "finish_reason": (
            final_generation_result.get(
                "finish_reason"
            )
        ),
    }

    # ---------------------------------------------
    # Final semantic validation
    # ---------------------------------------------

    final_validation = (
        final_semantic_result.get(
            "validation"
        )
        or {}
    )

    final_record[
        "validation"
    ] = {
        "status": "PASS",

        "protocol_version": (
            final_semantic_result.get(
                "validation_protocol_version"
            )
            or
            "golden_v2_qa_semantic_validation_v1"
        ),

        "structural_status": (
            final_semantic_result.get(
                "structural_status"
            )
        ),

        "semantic_verdict": (
            final_semantic_result.get(
                "semantic_verdict"
            )
        ),

        "question_targets_claim": (
            final_validation.get(
                "question_targets_claim"
            )
        ),

        "answer_recovers_claim": (
            final_validation.get(
                "answer_recovers_claim"
            )
        ),

        "scope_fidelity": (
            final_validation.get(
                "scope_fidelity"
            )
        ),

        "modality_quantifier_fidelity": (
            final_validation.get(
                "modality_quantifier_fidelity"
            )
        ),

        "standalone_question": (
            final_validation.get(
                "standalone_question"
            )
        ),

        "no_neighbor_fact_drift": (
            final_validation.get(
                "no_neighbor_fact_drift"
            )
        ),

        "no_external_knowledge": (
            final_validation.get(
                "no_external_knowledge"
            )
        ),

        "answer_question_alignment": (
            final_validation.get(
                "answer_question_alignment"
            )
        ),

        "failure_reasons": list(
            final_validation.get(
                "failure_reasons",
                [],
            )
            or []
        ),
    }

    # ---------------------------------------------
    # Final record metadata
    # ---------------------------------------------

    final_record[
        "golden_dataset_version"
    ] = (
        CELL57_DATASET_VERSION
    )

    final_record[
        "record_construction_version"
    ] = (
        CELL57_VERSION
    )

    final_record[
        "record_payload_sha256"
    ] = cell57_sha256_json(
        {
            key: value
            for key, value
            in final_record.items()
            if key
            != "record_payload_sha256"
        }
    )

    cell57_final_records.append(
        final_record
    )

    cell57_generation_source_counter[
        generation_source
    ] += 1


# =========================================================
# FINAL RECORD VALIDATION
# =========================================================

cell57_candidate_ids = [
    record["candidate_id"]
    for record
    in cell57_final_records
]


cell57_questions = [
    record[
        "evaluation_content"
    ][
        "question"
    ]
    for record
    in cell57_final_records
]


cell57_answers = [
    record[
        "evaluation_content"
    ][
        "expected_answer"
    ]
    for record
    in cell57_final_records
]


cell57_semantic_distribution = Counter(
    record[
        "validation"
    ][
        "semantic_verdict"
    ]
    for record
    in cell57_final_records
)


cell57_validation_dimension_fields = [
    "question_targets_claim",
    "answer_recovers_claim",
    "scope_fidelity",
    "modality_quantifier_fidelity",
    "standalone_question",
    "no_neighbor_fact_drift",
    "no_external_knowledge",
    "answer_question_alignment",
]


cell57_validation_dimension_failures = Counter()


for record in cell57_final_records:

    validation = record[
        "validation"
    ]

    for field_name in (
        cell57_validation_dimension_fields
    ):

        if (
            validation.get(
                field_name
            )
            is not True
        ):

            cell57_validation_dimension_failures[
                field_name
            ] += 1


# =========================================================
# DATASET CONTENT HASH
# =========================================================

cell57_dataset_payload_sha256 = (
    cell57_sha256_json(
        cell57_final_records
    )
)


# =========================================================
# HEALTH CHECKS
# =========================================================

cell57_checks = {
    "input_population_3813": (
        len(
            cell25_candidate_skeletons
        )
        == 3813
    ),

    "final_record_population_3813": (
        len(
            cell57_final_records
        )
        == 3813
    ),

    "zero_record_construction_errors": (
        len(
            cell57_record_errors
        )
        == 0
    ),

    "candidate_ids_unique": (
        len(
            set(
                cell57_candidate_ids
            )
        )
        == 3813
    ),

    "candidate_identity_exact": (
        set(
            cell57_candidate_ids
        )
        ==
        set(
            cell57_skeleton_candidate_ids
        )
    ),

    "all_questions_nonempty": (
        len(
            cell57_questions
        )
        == 3813
        and
        all(
            isinstance(
                value,
                str,
            )
            and value.strip()
            for value
            in cell57_questions
        )
    ),

    "all_answers_nonempty": (
        len(
            cell57_answers
        )
        == 3813
        and
        all(
            isinstance(
                value,
                str,
            )
            and value.strip()
            for value
            in cell57_answers
        )
    ),

    "semantic_pass_3813": (
        cell57_semantic_distribution
        ==
        Counter({
            "PASS": 3813
        })
    ),

    "all_validation_dimensions_true": (
        len(
            cell57_validation_dimension_failures
        )
        == 0
    ),

    "targeted_repair_population_37": (
        (
            cell57_generation_source_counter.get(
                "TARGETED_SEMANTIC_REPAIR",
                0,
            )
            +
            cell57_generation_source_counter.get(
                "FINAL_SINGLE_CANDIDATE_REPAIR",
                0,
            )
        )
        == 37
    ),

    "final_single_repair_population_1": (
        cell57_generation_source_counter.get(
            "FINAL_SINGLE_CANDIDATE_REPAIR",
            0,
        )
        == 1
    ),

    "zero_api_calls": True,
    "zero_file_writes": True,
}


cell57_status = (
    "PASS"
    if all(
        cell57_checks.values()
    )
    else "FAIL"
)


cell57_next_readiness = (
    "READY_FOR_GOLDEN_V2_FINAL_FREEZE"
    if cell57_status == "PASS"
    else
    "BLOCKED_CELL57_RECORD_CONSTRUCTION_REPAIR_REQUIRED"
)


# =========================================================
# OUTPUT
# =========================================================

print(
    "=== GOLDEN DATASET V2 FINAL "
    "RECORD CONSTRUCTION ==="
)

print()

print(
    "Input skeletons:",
    len(
        cell25_candidate_skeletons
    ),
)

print(
    "Final records:",
    len(
        cell57_final_records
    ),
)

print(
    "Construction errors:",
    len(
        cell57_record_errors
    ),
)

print()

print(
    "Generation source distribution:",
    dict(
        cell57_generation_source_counter
    ),
)

print(
    "Semantic distribution:",
    dict(
        cell57_semantic_distribution
    ),
)

print(
    "Validation dimension failures:",
    dict(
        cell57_validation_dimension_failures
    ),
)

print()

print(
    "Dataset payload SHA256:",
    cell57_dataset_payload_sha256,
)

print()

if cell57_record_errors:

    print(
        "Record construction error examples:"
    )

    for item in (
        cell57_record_errors[:10]
    ):

        print(
            " -",
            item,
        )

    print()

print("Checks:")

for (
    check_name,
    check_value,
) in cell57_checks.items():

    print(
        " -",
        check_name,
        ":",
        (
            "PASS"
            if check_value
            else "FAIL"
        ),
    )

print()

print(
    "CELL 57 STATUS:",
    cell57_status,
)

print(
    "Next readiness:",
    cell57_next_readiness,
)


assert cell57_status == "PASS", (
    "Cell 57 final record construction failed."
)
````

<h2>مرحله ۵۸ — Freeze و ثبت Artifactهای نهایی Golden Dataset v2</h2>
<p>
ساخت رکوردهای نهایی با موفقیت کامل انجام شد: ۳۸۱۳ رکورد بدون خطای ساخت، با هویت یکتا، Question و Expected Answer معتبر و Semantic PASS کامل آماده هستند. Hash محتوای قطعی این جمعیت نیز در مرحله قبل محاسبه شده است.
</p>
<p>
در این مرحله Golden Dataset v2 برای نخستین بار به‌عنوان Artifact نهایی روی دیسک Freeze می‌شود. فایل اصلی Dataset به‌صورت JSONL نوشته خواهد شد و یک Manifest و Final Report نیز شامل نسخه Dataset، تعداد رکوردها، توزیع مسیر تولید، وضعیت Validation و Hashهای رمزنگاری‌شده ساخته می‌شوند.
</p>
<p>
Freeze به‌صورت محافظه‌کارانه انجام می‌شود: اگر Artifact نهایی از قبل وجود داشته باشد، فقط در صورتی پذیرفته می‌شود که محتوای آن دقیقاً با خروجی قطعی فعلی یکسان باشد؛ محتوای متفاوت به‌صورت خودکار Overwrite نخواهد شد. پس از نوشتن، فایل‌ها دوباره از دیسک خوانده می‌شوند و تعداد رکوردها، هویت Candidateها و SHA256 آنها مستقل از وضعیت حافظه بررسی می‌شود. این مرحله هیچ فراخوانی به API یا مدل ندارد.
</p>
</div>

````python
# Code Cell 58 — Freeze Final Golden Dataset v2 Artifacts

from collections import Counter
from pathlib import Path
import hashlib
import json
import os
import tempfile


# =========================================================
# CONFIG
# =========================================================

CELL58_FREEZE_VERSION = (
    "golden_dataset_v2_final_freeze_v1"
)

CELL58_DATASET_FILENAME = (
    "golden_dataset_v2.jsonl"
)

CELL58_MANIFEST_FILENAME = (
    "golden_dataset_v2_manifest.json"
)

CELL58_REPORT_FILENAME = (
    "golden_dataset_v2_final_report.json"
)


# =========================================================
# REQUIRED STATE
# =========================================================

cell58_required_names = [
    "project_root",
    "cell57_final_records",
    "cell57_dataset_payload_sha256",
    "cell57_generation_source_counter",
    "CELL57_DATASET_VERSION",
    "CELL57_VERSION",
]

cell58_missing_state = [
    name
    for name in cell58_required_names
    if name not in globals()
]

assert not cell58_missing_state, (
    "Missing required Cell 58 state: "
    + ", ".join(cell58_missing_state)
)


assert len(
    cell57_final_records
) == 3813


# =========================================================
# PATHS
# =========================================================

cell58_output_dir = (
    Path(project_root)
    / "artifacts"
    / "phase2"
    / "golden_dataset_v2"
)

cell58_output_dir.mkdir(
    parents=True,
    exist_ok=True,
)


cell58_dataset_path = (
    cell58_output_dir
    / CELL58_DATASET_FILENAME
)

cell58_manifest_path = (
    cell58_output_dir
    / CELL58_MANIFEST_FILENAME
)

cell58_report_path = (
    cell58_output_dir
    / CELL58_REPORT_FILENAME
)


# =========================================================
# CANONICAL SERIALIZATION
# =========================================================

def cell58_canonical_json(value):

    return json.dumps(
        value,
        ensure_ascii=False,
        sort_keys=True,
        separators=(",", ":"),
    )


def cell58_sha256_bytes(data):

    return hashlib.sha256(
        data
    ).hexdigest()


def cell58_sha256_file(path):

    digest = hashlib.sha256()

    with Path(path).open(
        "rb"
    ) as handle:

        for chunk in iter(
            lambda: handle.read(
                1024 * 1024
            ),
            b"",
        ):

            digest.update(chunk)

    return digest.hexdigest()


# =========================================================
# EXACT DATASET BYTES
# =========================================================

cell58_dataset_text = (
    "\n".join(
        cell58_canonical_json(
            record
        )
        for record
        in cell57_final_records
    )
    + "\n"
)

cell58_dataset_bytes = (
    cell58_dataset_text.encode(
        "utf-8"
    )
)

CELL58_EXPECTED_DATASET_FILE_SHA256 = (
    cell58_sha256_bytes(
        cell58_dataset_bytes
    )
)


# =========================================================
# DATASET CONTENT IDENTITY
# =========================================================

cell58_recomputed_payload_sha256 = (
    hashlib.sha256(
        cell58_canonical_json(
            cell57_final_records
        ).encode("utf-8")
    ).hexdigest()
)

assert (
    cell58_recomputed_payload_sha256
    ==
    cell57_dataset_payload_sha256
), (
    "Cell 57 dataset payload hash changed "
    "before final freeze."
)


# =========================================================
# FINAL SUMMARY INPUTS
# =========================================================

cell58_generation_distribution = dict(
    sorted(
        cell57_generation_source_counter.items()
    )
)

cell58_semantic_distribution = {
    "PASS": 3813,
    "FAIL": 0,
}

cell58_structural_distribution = {
    "PASS": 3813,
    "FAIL": 0,
}


# =========================================================
# MANIFEST PAYLOAD
# =========================================================

cell58_manifest_payload = {
    "artifact_type": (
        "golden_dataset_v2_manifest"
    ),

    "freeze_version": (
        CELL58_FREEZE_VERSION
    ),

    "dataset_version": (
        CELL57_DATASET_VERSION
    ),

    "record_construction_version": (
        CELL57_VERSION
    ),

    "dataset_filename": (
        CELL58_DATASET_FILENAME
    ),

    "record_count": 3813,

    "dataset_payload_sha256": (
        cell57_dataset_payload_sha256
    ),

    "dataset_file_sha256": (
        CELL58_EXPECTED_DATASET_FILE_SHA256
    ),

    "generation_source_distribution": (
        cell58_generation_distribution
    ),

    "structural_distribution": (
        cell58_structural_distribution
    ),

    "semantic_distribution": (
        cell58_semantic_distribution
    ),

    "api_failures": 0,

    "semantic_failures": 0,

    "freeze_status": (
        "FROZEN"
    ),
}


# =========================================================
# FINAL REPORT PAYLOAD
# =========================================================

cell58_report_payload = {
    "report_type": (
        "golden_dataset_v2_final_report"
    ),

    "freeze_version": (
        CELL58_FREEZE_VERSION
    ),

    "dataset_version": (
        CELL57_DATASET_VERSION
    ),

    "dataset_path": (
        str(
            cell58_dataset_path.relative_to(
                project_root
            )
        )
    ),

    "manifest_path": (
        str(
            cell58_manifest_path.relative_to(
                project_root
            )
        )
    ),

    "record_count": 3813,

    "generation": {
        "full_generation_v4": (
            cell58_generation_distribution.get(
                "FULL_GENERATION_V4",
                0,
            )
        ),

        "targeted_semantic_repair": (
            cell58_generation_distribution.get(
                "TARGETED_SEMANTIC_REPAIR",
                0,
            )
        ),

        "final_single_candidate_repair": (
            cell58_generation_distribution.get(
                "FINAL_SINGLE_CANDIDATE_REPAIR",
                0,
            )
        ),
    },

    "validation": {
        "structural_pass": 3813,
        "structural_fail": 0,
        "semantic_pass": 3813,
        "semantic_fail": 0,
        "api_failures": 0,
    },

    "hashes": {
        "dataset_payload_sha256": (
            cell57_dataset_payload_sha256
        ),

        "dataset_file_sha256": (
            CELL58_EXPECTED_DATASET_FILE_SHA256
        ),
    },

    "status": (
        "PASS"
    ),

    "freeze_status": (
        "FROZEN"
    ),
}


# =========================================================
# EXACT JSON ARTIFACT BYTES
# =========================================================

cell58_manifest_text = (
    json.dumps(
        cell58_manifest_payload,
        ensure_ascii=False,
        sort_keys=True,
        indent=2,
    )
    + "\n"
)

cell58_manifest_bytes = (
    cell58_manifest_text.encode(
        "utf-8"
    )
)


cell58_report_text = (
    json.dumps(
        cell58_report_payload,
        ensure_ascii=False,
        sort_keys=True,
        indent=2,
    )
    + "\n"
)

cell58_report_bytes = (
    cell58_report_text.encode(
        "utf-8"
    )
)


CELL58_EXPECTED_MANIFEST_SHA256 = (
    cell58_sha256_bytes(
        cell58_manifest_bytes
    )
)

CELL58_EXPECTED_REPORT_SHA256 = (
    cell58_sha256_bytes(
        cell58_report_bytes
    )
)


# =========================================================
# SAFE FREEZE WRITE
# =========================================================

def cell58_atomic_write_bytes(
    path,
    data,
):

    path = Path(path)

    fd, temp_name = tempfile.mkstemp(
        prefix=path.name + ".",
        suffix=".tmp",
        dir=str(path.parent),
    )

    try:

        with os.fdopen(
            fd,
            "wb",
        ) as handle:

            handle.write(data)
            handle.flush()
            os.fsync(
                handle.fileno()
            )

        os.replace(
            temp_name,
            path,
        )

    except Exception:

        try:
            os.unlink(
                temp_name
            )
        except FileNotFoundError:
            pass

        raise


def cell58_freeze_exact_artifact(
    path,
    expected_bytes,
):

    path = Path(path)

    expected_sha256 = (
        cell58_sha256_bytes(
            expected_bytes
        )
    )

    if path.exists():

        existing_sha256 = (
            cell58_sha256_file(
                path
            )
        )

        if (
            existing_sha256
            != expected_sha256
        ):

            raise AssertionError(
                "Frozen artifact already exists "
                "with different content; refusing "
                f"overwrite: {path}"
            )

        return {
            "action": (
                "EXISTING_EXACT_ARTIFACT_ACCEPTED"
            ),

            "sha256": (
                existing_sha256
            ),
        }

    cell58_atomic_write_bytes(
        path,
        expected_bytes,
    )

    written_sha256 = (
        cell58_sha256_file(
            path
        )
    )

    assert (
        written_sha256
        == expected_sha256
    )

    return {
        "action": (
            "NEW_FROZEN_ARTIFACT_WRITTEN"
        ),

        "sha256": (
            written_sha256
        ),
    }


# =========================================================
# FREEZE
# =========================================================

cell58_dataset_freeze = (
    cell58_freeze_exact_artifact(
        cell58_dataset_path,
        cell58_dataset_bytes,
    )
)

cell58_manifest_freeze = (
    cell58_freeze_exact_artifact(
        cell58_manifest_path,
        cell58_manifest_bytes,
    )
)

cell58_report_freeze = (
    cell58_freeze_exact_artifact(
        cell58_report_path,
        cell58_report_bytes,
    )
)


# =========================================================
# DISK RELOAD — DATASET
# =========================================================

cell58_disk_records = []

with cell58_dataset_path.open(
    "r",
    encoding="utf-8",
) as handle:

    for line_number, line in enumerate(
        handle,
        start=1,
    ):

        stripped = line.strip()

        if not stripped:
            continue

        try:

            record = json.loads(
                stripped
            )

        except Exception as exc:

            raise AssertionError(
                "Invalid JSONL at line "
                f"{line_number}: "
                f"{type(exc).__name__}: {exc}"
            )

        cell58_disk_records.append(
            record
        )


# =========================================================
# DISK RELOAD — MANIFEST / REPORT
# =========================================================

with cell58_manifest_path.open(
    "r",
    encoding="utf-8",
) as handle:

    cell58_disk_manifest = (
        json.load(handle)
    )


with cell58_report_path.open(
    "r",
    encoding="utf-8",
) as handle:

    cell58_disk_report = (
        json.load(handle)
    )


# =========================================================
# DISK HASHES
# =========================================================

CELL58_DATASET_FILE_SHA256 = (
    cell58_sha256_file(
        cell58_dataset_path
    )
)

CELL58_MANIFEST_FILE_SHA256 = (
    cell58_sha256_file(
        cell58_manifest_path
    )
)

CELL58_REPORT_FILE_SHA256 = (
    cell58_sha256_file(
        cell58_report_path
    )
)


cell58_disk_payload_sha256 = (
    hashlib.sha256(
        cell58_canonical_json(
            cell58_disk_records
        ).encode("utf-8")
    ).hexdigest()
)


# =========================================================
# DISK IDENTITY
# =========================================================

cell58_disk_candidate_ids = [
    record["candidate_id"]
    for record
    in cell58_disk_records
]


cell58_memory_candidate_ids = [
    record["candidate_id"]
    for record
    in cell57_final_records
]


# =========================================================
# FINAL FREEZE HEALTH
# =========================================================

cell58_checks = {
    "disk_record_count_3813": (
        len(
            cell58_disk_records
        )
        == 3813
    ),

    "disk_candidate_ids_unique": (
        len(
            set(
                cell58_disk_candidate_ids
            )
        )
        == 3813
    ),

    "disk_candidate_identity_exact": (
        cell58_disk_candidate_ids
        ==
        cell58_memory_candidate_ids
    ),

    "disk_payload_hash_matches_cell57": (
        cell58_disk_payload_sha256
        ==
        cell57_dataset_payload_sha256
    ),

    "dataset_file_hash_exact": (
        CELL58_DATASET_FILE_SHA256
        ==
        CELL58_EXPECTED_DATASET_FILE_SHA256
    ),

    "manifest_file_hash_exact": (
        CELL58_MANIFEST_FILE_SHA256
        ==
        CELL58_EXPECTED_MANIFEST_SHA256
    ),

    "report_file_hash_exact": (
        CELL58_REPORT_FILE_SHA256
        ==
        CELL58_EXPECTED_REPORT_SHA256
    ),

    "manifest_record_count_3813": (
        cell58_disk_manifest.get(
            "record_count"
        )
        == 3813
    ),

    "manifest_payload_hash_exact": (
        cell58_disk_manifest.get(
            "dataset_payload_sha256"
        )
        ==
        cell57_dataset_payload_sha256
    ),

    "manifest_file_hash_exact_reference": (
        cell58_disk_manifest.get(
            "dataset_file_sha256"
        )
        ==
        CELL58_DATASET_FILE_SHA256
    ),

    "manifest_frozen": (
        cell58_disk_manifest.get(
            "freeze_status"
        )
        == "FROZEN"
    ),

    "report_record_count_3813": (
        cell58_disk_report.get(
            "record_count"
        )
        == 3813
    ),

    "report_semantic_pass_3813": (
        cell58_disk_report.get(
            "validation",
            {},
        ).get(
            "semantic_pass"
        )
        == 3813
    ),

    "report_semantic_fail_0": (
        cell58_disk_report.get(
            "validation",
            {},
        ).get(
            "semantic_fail"
        )
        == 0
    ),

    "report_api_failures_0": (
        cell58_disk_report.get(
            "validation",
            {},
        ).get(
            "api_failures"
        )
        == 0
    ),

    "zero_api_calls": True,
}


cell58_status = (
    "PASS"
    if all(
        cell58_checks.values()
    )
    else "FAIL"
)


cell58_next_readiness = (
    "READY_FOR_PAID_CELL_RECONSTRUCTION_AUDIT"
    if cell58_status == "PASS"
    else
    "BLOCKED_CELL58_FREEZE_REPAIR_REQUIRED"
)


# =========================================================
# OUTPUT
# =========================================================

print(
    "=== GOLDEN DATASET V2 FINAL FREEZE ==="
)

print()

print(
    "Dataset:",
    str(
        cell58_dataset_path.relative_to(
            project_root
        )
    ),
)

print(
    "Dataset freeze action:",
    cell58_dataset_freeze[
        "action"
    ],
)

print(
    "Dataset records:",
    len(
        cell58_disk_records
    ),
)

print(
    "Dataset payload SHA256:",
    cell58_disk_payload_sha256,
)

print(
    "Dataset file SHA256:",
    CELL58_DATASET_FILE_SHA256,
)

print()

print(
    "Manifest:",
    str(
        cell58_manifest_path.relative_to(
            project_root
        )
    ),
)

print(
    "Manifest freeze action:",
    cell58_manifest_freeze[
        "action"
    ],
)

print(
    "Manifest SHA256:",
    CELL58_MANIFEST_FILE_SHA256,
)

print()

print(
    "Final report:",
    str(
        cell58_report_path.relative_to(
            project_root
        )
    ),
)

print(
    "Final report freeze action:",
    cell58_report_freeze[
        "action"
    ],
)

print(
    "Final report SHA256:",
    CELL58_REPORT_FILE_SHA256,
)

print()

print(
    "Generation source distribution:",
    cell58_generation_distribution,
)

print(
    "Structural distribution:",
    cell58_structural_distribution,
)

print(
    "Semantic distribution:",
    cell58_semantic_distribution,
)

print()

print("Freeze checks:")

for (
    check_name,
    check_value,
) in cell58_checks.items():

    print(
        " -",
        check_name,
        ":",
        (
            "PASS"
            if check_value
            else "FAIL"
        ),
    )

print()

print(
    "CELL 58 STATUS:",
    cell58_status,
)

print(
    "Next readiness:",
    cell58_next_readiness,
)


assert cell58_status == "PASS", (
    "Cell 58 final Golden Dataset v2 freeze failed."
)
````

<h2>مرحله ۵۹ — ممیزی Cellهای دارای قابلیت اجرای API</h2>
<p>
Golden Dataset v2 با ۳۸۱۳ رکورد نهایی Freeze شده و Dataset، Manifest و Final Report همگی از روی دیسک با Hashهای مورد انتظار تأیید شده‌اند. از این نقطه محتوای Dataset نهایی تغییر نخواهد کرد.
</p>
<p>
پیش از Restart Kernel → Run All باید تضمین شود که هیچ Cell تاریخی یا جدیدی دوباره درخواست پولی به مدل ارسال نمی‌کند. در این مرحله خود فایل Notebook به‌صورت Read-only بررسی می‌شود و تمام Code Cellهایی که دارای فراخوانی chat.completions.create، فلگ‌های اجازه API یا سایر نشانه‌های اجرای مدل هستند شناسایی می‌شوند.
</p>
<p>
این مرحله هیچ Notebook Cell، Artifact یا Checkpointی را تغییر نمی‌دهد و هیچ API Call انجام نمی‌دهد. خروجی آن فهرست دقیق Cellهای نیازمند تبدیل به حالت Reconstruction-only است تا در مراحل بعد فقط همان Cellها اصلاح شوند و سپس Restart → Run All کاملاً بدون مصرف API انجام شود.
</p>
</div>

````python
# Code Cell 59 — Read-Only Paid/API Cell Reconstruction Audit

from pathlib import Path
from collections import Counter
import json
import re


# =========================================================
# CONFIG
# =========================================================

CELL59_VERSION = (
    "golden_v2_paid_api_cell_reconstruction_audit_v2"
)

CELL59_EXPECTED_NOTEBOOK_NAME = (
    "12j_golden_dataset_v2_registry_driven_rebuild_RUN_ALL_SAFE.ipynb"
)


# =========================================================
# LOCATE EXACT RUN-ALL-SAFE NOTEBOOK
# =========================================================

cell59_project_root = Path(
    project_root
).resolve()


cell59_exact_candidates = sorted(
    path
    for path
    in cell59_project_root.rglob(
        CELL59_EXPECTED_NOTEBOOK_NAME
    )
)


assert len(
    cell59_exact_candidates
) == 1, (
    "Expected exactly one RUN_ALL_SAFE NB12j notebook. Found: "
    + repr(
        [
            str(
                path.relative_to(
                    cell59_project_root
                )
            )
            for path
            in cell59_exact_candidates
        ]
    )
)


cell59_notebook_path = (
    cell59_exact_candidates[0]
)


# =========================================================
# READ NOTEBOOK — NO WRITE
# =========================================================

with cell59_notebook_path.open(
    "r",
    encoding="utf-8",
) as handle:

    cell59_notebook = json.load(
        handle
    )


cell59_cells = (
    cell59_notebook.get(
        "cells",
        []
    )
)


# =========================================================
# HELPERS
# =========================================================

def cell59_source_text(cell):

    source = cell.get(
        "source",
        []
    )

    if isinstance(
        source,
        list,
    ):
        return "".join(source)

    if isinstance(
        source,
        str,
    ):
        return source

    return ""


def cell59_extract_code_cell_number(
    source_text,
):

    match = re.search(
        r"^\s*#\s*Code\s+Cell\s+(\d+)\b",
        source_text,
        flags=(
            re.IGNORECASE
            | re.MULTILINE
        ),
    )

    if match:
        return int(
            match.group(1)
        )

    return None


# =========================================================
# API / PAID EXECUTION SIGNALS
# =========================================================

cell59_signal_patterns = {
    "chat_completions_create": (
        r"\.chat\s*\.completions\s*\.create\s*\("
    ),

    "completions_create": (
        r"\.completions\s*\.create\s*\("
    ),

    "allow_api_true": (
        r"\b[A-Z0-9_]*ALLOW_API[A-Z0-9_]*"
        r"\s*=\s*True\b"
    ),

    "execute_full_review_true": (
        r"\bEXECUTE_FULL_REVIEW\s*=\s*True\b"
    ),

    "api_for_missing_true": (
        r"\b[A-Z0-9_]*API[A-Z0-9_]*MISSING[A-Z0-9_]*"
        r"\s*=\s*True\b"
    ),

    "openai_client_call": (
        r"\bsemantic_review_client\b"
    ),
}


# =========================================================
# SCAN CODE CELLS
# =========================================================

cell59_code_cells = []
cell59_paid_capable_cells = []


for notebook_index, cell in enumerate(
    cell59_cells
):

    if (
        cell.get(
            "cell_type"
        )
        != "code"
    ):
        continue

    source_text = (
        cell59_source_text(
            cell
        )
    )

    code_cell_number = (
        cell59_extract_code_cell_number(
            source_text
        )
    )

    signals = []

    for (
        signal_name,
        pattern,
    ) in cell59_signal_patterns.items():

        if re.search(
            pattern,
            source_text,
            flags=(
                re.IGNORECASE
                | re.MULTILINE
            ),
        ):

            signals.append(
                signal_name
            )

    record = {
        "notebook_index": (
            notebook_index
        ),

        "code_cell_number": (
            code_cell_number
        ),

        "signals": sorted(
            set(signals)
        ),

        "source_first_line": (
            source_text
            .strip()
            .splitlines()[0]
            if source_text.strip()
            else ""
        ),
    }

    cell59_code_cells.append(
        record
    )

    paid_execution_signals = {
        "allow_api_true",
        "execute_full_review_true",
        "api_for_missing_true",
    }

    if (
        set(
            record["signals"]
        )
        &
        paid_execution_signals
    ):

        cell59_paid_capable_cells.append(
            record
        )


# =========================================================
# CELL NUMBER HEALTH
# =========================================================

cell59_numbered_code_cells = [
    record
    for record
    in cell59_code_cells
    if record[
        "code_cell_number"
    ]
    is not None
]


cell59_code_cell_numbers = [
    record[
        "code_cell_number"
    ]
    for record
    in cell59_numbered_code_cells
]


cell59_number_counter = Counter(
    cell59_code_cell_numbers
)


cell59_duplicate_code_cell_numbers = sorted(
    number
    for number, count
    in cell59_number_counter.items()
    if count > 1
)


cell59_expected_numbers = list(
    range(
        1,
        61,
    )
)


# =========================================================
# DIRECT API DEFINITIONS VS ENABLED EXECUTION
# =========================================================

cell59_direct_api_definition_cells = sorted({
    record["code_cell_number"]
    for record
    in cell59_code_cells
    if (
        record[
            "code_cell_number"
        ]
        is not None
        and
        (
            "chat_completions_create"
            in record["signals"]
            or
            "completions_create"
            in record["signals"]
        )
    )
})


cell59_api_enabled_cells = sorted({
    record["code_cell_number"]
    for record
    in cell59_code_cells
    if (
        record[
            "code_cell_number"
        ]
        is not None
        and
        (
            "allow_api_true"
            in record["signals"]
            or
            "api_for_missing_true"
            in record["signals"]
            or
            "execute_full_review_true"
            in record["signals"]
        )
    )
})


# =========================================================
# RUN-ALL SAFETY HEALTH
# =========================================================

cell59_checks = {
    "exact_run_all_safe_notebook_found": (
        len(
            cell59_exact_candidates
        )
        == 1
    ),

    "notebook_has_cells": (
        len(
            cell59_cells
        )
        > 0
    ),

    "numbered_code_cells_60": (
        len(
            cell59_numbered_code_cells
        )
        == 60
    ),

    "code_cell_numbers_exact_1_to_60": (
        sorted(
            cell59_code_cell_numbers
        )
        == cell59_expected_numbers
    ),

    "no_duplicate_code_cell_numbers": (
        len(
            cell59_duplicate_code_cell_numbers
        )
        == 0
    ),

    "zero_api_enabled_cells": (
        len(
            cell59_api_enabled_cells
        )
        == 0
    ),

    "zero_api_calls_this_cell": True,

    "zero_file_writes_this_cell": True,
}


cell59_status = (
    "PASS"
    if all(
        cell59_checks.values()
    )
    else "FAIL"
)


cell59_next_readiness = (
    "READY_TO_CONTINUE_FINAL_RUN_ALL"
    if cell59_status == "PASS"
    else
    "BLOCKED_CELL59_RUN_ALL_SAFETY_REPAIR_REQUIRED"
)


# =========================================================
# OUTPUT
# =========================================================

print(
    "=== GOLDEN V2 FINAL RUN-ALL API SAFETY AUDIT ==="
)

print()

print(
    "Notebook:",
    str(
        cell59_notebook_path.relative_to(
            cell59_project_root
        )
    ),
)

print(
    "Notebook cells:",
    len(
        cell59_cells
    ),
)

print(
    "Numbered code cells:",
    len(
        cell59_numbered_code_cells
    ),
)

print()

print(
    "Direct API-definition cells:",
    cell59_direct_api_definition_cells,
)

print(
    "API-enabled cells:",
    cell59_api_enabled_cells,
)

print()

print(
    "Duplicate code-cell numbers:",
    cell59_duplicate_code_cell_numbers,
)

print()

print("Run-All safety checks:")

for (
    check_name,
    check_value,
) in cell59_checks.items():

    print(
        " -",
        check_name,
        ":",
        (
            "PASS"
            if check_value
            else "FAIL"
        ),
    )

print()

print(
    "CELL 59 STATUS:",
    cell59_status,
)

print(
    "Next readiness:",
    cell59_next_readiness,
)


assert cell59_status == "PASS", (
    "Cell 59 final Run-All API safety audit failed."
)
````

<h2>مرحله ۶۰ — ساخت نقشه دقیق تبدیل Cellهای پولی به Reconstruction-only</h2>
<p>
ممیزی مرحله ۵۹ نشان داد Cellهای ۲۷، ۳۱، ۳۳، ۳۵، ۳۶، ۳۹، ۴۲، ۴۳، ۴۴، ۴۶ و ۴۷ هنوز دارای کد فراخوانی مستقیم API هستند. همچنین Cell ۴۸ نیز دارای قابلیت API است، اما به دلیل نبود علامت # در ابتدای Header، شماره آن توسط ممیزی قبلی استخراج نشده است.
</p>
<p>
در این مرحله هیچ Cell یا فایلی تغییر نمی‌کند. Source واقعی همین Cellهای Notebook بررسی می‌شود تا فلگ‌های اجرای API، مسیرهای Checkpoint، الگوی Load/Recovery و وضعیت Header هر Cell استخراج شود. هدف این است که تبدیل Reconstruction-only بر اساس کد واقعی Notebook انجام شود و هیچ Cell تاریخی با حدس بازنویسی نشود.
</p>
<p>
این مرحله هیچ API Call و هیچ File Write ندارد. خروجی آن نقشه قطعی Cellهایی است که باید پیش از Restart Kernel → Run All اصلاح شوند.
</p>

````python
# Code Cell 60 — Exact Reconstruction-Only Conversion Map

from pathlib import Path
import json
import re


# =========================================================
# CONFIG
# =========================================================

CELL60_VERSION = (
    "golden_v2_reconstruction_only_conversion_map_v1"
)


# =========================================================
# REQUIRED STATE
# =========================================================

assert "project_root" in globals()


cell60_project_root = Path(
    project_root
).resolve()


# =========================================================
# LOCATE NB12j
# =========================================================

CELL60_EXPECTED_NOTEBOOK_NAME = (
    "12j_golden_dataset_v2_registry_driven_rebuild_RUN_ALL_SAFE.ipynb"
)

cell60_candidates = sorted(
    cell60_project_root.rglob(
        CELL60_EXPECTED_NOTEBOOK_NAME
    )
)

assert len(cell60_candidates) == 1, (
    "Expected exactly one RUN_ALL_SAFE NB12j notebook. Found: "
    + repr(
        [
            str(
                path.relative_to(
                    cell60_project_root
                )
            )
            for path
            in cell60_candidates
        ]
    )
)

cell60_notebook_path = (
    cell60_candidates[0]
)


with cell60_notebook_path.open(
    "r",
    encoding="utf-8",
) as handle:

    cell60_notebook = json.load(handle)


# =========================================================
# HELPERS
# =========================================================

def cell60_source_text(cell):

    source = cell.get(
        "source",
        []
    )

    if isinstance(source, list):
        return "".join(source)

    if isinstance(source, str):
        return source

    return ""


def cell60_extract_number(source_text):

    match = re.search(
        r"^\s*#?\s*Code\s+Cell\s+(\d+)\b",
        source_text,
        flags=re.IGNORECASE
        | re.MULTILINE,
    )

    if match:
        return int(match.group(1))

    return None


# =========================================================
# TARGET CELLS
# =========================================================

CELL60_TARGET_NUMBERS = {
    27,
    31,
    33,
    35,
    36,
    39,
    42,
    43,
    44,
    46,
    47,
    48,
}


cell60_target_records = []


for notebook_index, cell in enumerate(
    cell60_notebook.get(
        "cells",
        []
    )
):

    if cell.get("cell_type") != "code":
        continue

    source_text = cell60_source_text(
        cell
    )

    cell_number = (
        cell60_extract_number(
            source_text
        )
    )

    if cell_number not in (
        CELL60_TARGET_NUMBERS
    ):
        continue

    # ---------------------------------------------
    # API flags
    # ---------------------------------------------

    api_flags = []

    for match in re.finditer(
        r"(?m)^\s*"
        r"([A-Z][A-Z0-9_]*(?:ALLOW_API|API_FOR_MISSING)"
        r"[A-Z0-9_]*)"
        r"\s*=\s*(True|False)\s*$",
        source_text,
    ):

        api_flags.append({
            "name": match.group(1),
            "value": match.group(2),
        })

    # ---------------------------------------------
    # Other execution flags
    # ---------------------------------------------

    execution_flags = []

    for match in re.finditer(
        r"(?m)^\s*"
        r"(EXECUTE_[A-Z0-9_]+)"
        r"\s*=\s*(True|False)\s*$",
        source_text,
    ):

        execution_flags.append({
            "name": match.group(1),
            "value": match.group(2),
        })

    # ---------------------------------------------
    # Checkpoint path signals
    # ---------------------------------------------

    checkpoint_literals = sorted(
        set(
            re.findall(
                r'["\']([^"\']*'
                r'(?:checkpoint|work)'
                r'[^"\']*)["\']',
                source_text,
                flags=re.IGNORECASE,
            )
        )
    )

    # ---------------------------------------------
    # Functional signals
    # ---------------------------------------------

    has_direct_api_call = bool(
        re.search(
            r"\.chat\s*\.completions"
            r"\s*\.create\s*\(",
            source_text,
            flags=re.IGNORECASE,
        )
    )

    has_checkpoint_read = bool(
        re.search(
            r"\.(?:open|read_text)\s*\("
            r"|json\.load\s*\(",
            source_text,
            flags=re.IGNORECASE,
        )
    )

    has_checkpoint_write = bool(
        re.search(
            r"(?:write_text|json\.dump|os\.replace"
            r"|atomic_write|checkpoint.*write)",
            source_text,
            flags=re.IGNORECASE,
        )
    )

    has_missing_branch = bool(
        re.search(
            r"\bMISSING\b"
            r"|missing.*checkpoint"
            r"|checkpoint.*missing",
            source_text,
            flags=re.IGNORECASE,
        )
    )

    first_line = (
        source_text.strip().splitlines()[0]
        if source_text.strip()
        else ""
    )

    header_has_hash = bool(
        re.match(
            r"^\s*#",
            first_line,
        )
    )

    cell60_target_records.append({
        "cell_number": cell_number,
        "notebook_index": notebook_index,
        "first_line": first_line,
        "header_has_hash": header_has_hash,
        "direct_api_call": has_direct_api_call,
        "api_flags": api_flags,
        "execution_flags": execution_flags,
        "checkpoint_read_signal": (
            has_checkpoint_read
        ),
        "checkpoint_write_signal": (
            has_checkpoint_write
        ),
        "missing_checkpoint_branch": (
            has_missing_branch
        ),
        "checkpoint_literals": (
            checkpoint_literals
        ),
        "source_line_count": len(
            source_text.splitlines()
        ),
    })


# =========================================================
# EXACT COVERAGE
# =========================================================

cell60_found_numbers = {
    record["cell_number"]
    for record
    in cell60_target_records
}


cell60_missing_target_numbers = sorted(
    CELL60_TARGET_NUMBERS
    - cell60_found_numbers
)


cell60_duplicate_numbers = sorted({
    number
    for number in cell60_found_numbers
    if sum(
        1
        for record
        in cell60_target_records
        if record["cell_number"] == number
    )
    > 1
})


# =========================================================
# CLASSIFY REQUIRED CONVERSION
# =========================================================

cell60_conversion_plan = []


for record in sorted(
    cell60_target_records,
    key=lambda item: item["cell_number"],
):

    true_api_flags = [
        flag["name"]
        for flag
        in record["api_flags"]
        if flag["value"] == "True"
    ]

    false_api_flags = [
        flag["name"]
        for flag
        in record["api_flags"]
        if flag["value"] == "False"
    ]

    if true_api_flags:

        conversion_type = (
            "DISABLE_API_FLAG_AND_VERIFY_"
            "CHECKPOINT_RECONSTRUCTION"
        )

    elif record["direct_api_call"]:

        conversion_type = (
            "STRUCTURAL_REWRITE_TO_"
            "CHECKPOINT_RECONSTRUCTION_ONLY"
        )

    else:

        conversion_type = (
            "VERIFY_ONLY"
        )

    if not record["header_has_hash"]:

        conversion_type += (
            "_AND_FIX_HEADER"
        )

    cell60_conversion_plan.append({
        **record,
        "true_api_flags": true_api_flags,
        "false_api_flags": false_api_flags,
        "conversion_type": conversion_type,
    })


# =========================================================
# SAFETY CHECKS
# =========================================================

cell60_checks = {
    "all_12_target_cells_found": (
        len(cell60_target_records) == 12
    ),

    "exact_target_number_coverage": (
        cell60_found_numbers
        == CELL60_TARGET_NUMBERS
    ),

    "zero_missing_targets": (
        len(
            cell60_missing_target_numbers
        )
        == 0
    ),

    "zero_duplicate_targets": (
        len(
            cell60_duplicate_numbers
        )
        == 0
    ),

    "cell48_found": (
        48 in cell60_found_numbers
    ),

    "zero_api_calls": True,

    "zero_file_writes": True,
}


cell60_status = (
    "PASS"
    if all(
        cell60_checks.values()
    )
    else "FAIL"
)


cell60_next_readiness = (
    "READY_FOR_EXACT_HISTORICAL_CELL_CONVERSION"
    if cell60_status == "PASS"
    else
    "BLOCKED_CELL60_AUDIT_REPAIR_REQUIRED"
)


# =========================================================
# OUTPUT
# =========================================================

print(
    "=== GOLDEN V2 RECONSTRUCTION-ONLY "
    "CONVERSION MAP ==="
)

print()

print(
    "Notebook:",
    str(
        cell60_notebook_path.relative_to(
            cell60_project_root
        )
    ),
)

print(
    "Target cells:",
    sorted(
        CELL60_TARGET_NUMBERS
    ),
)

print(
    "Found cells:",
    sorted(
        cell60_found_numbers
    ),
)

print()

for record in (
    cell60_conversion_plan
):

    print(
        f"--- Code Cell "
        f"{record['cell_number']} ---"
    )

    print(
        "Notebook index:",
        record[
            "notebook_index"
        ],
    )

    print(
        "Header:",
        record[
            "first_line"
        ],
    )

    print(
        "Header has #:",
        record[
            "header_has_hash"
        ],
    )

    print(
        "Direct API call:",
        record[
            "direct_api_call"
        ],
    )

    print(
        "True API flags:",
        record[
            "true_api_flags"
        ],
    )

    print(
        "False API flags:",
        record[
            "false_api_flags"
        ],
    )

    print(
        "Checkpoint read signal:",
        record[
            "checkpoint_read_signal"
        ],
    )

    print(
        "Checkpoint write signal:",
        record[
            "checkpoint_write_signal"
        ],
    )

    print(
        "Missing-checkpoint branch:",
        record[
            "missing_checkpoint_branch"
        ],
    )

    print(
        "Checkpoint literals:",
        record[
            "checkpoint_literals"
        ][:10],
    )

    print(
        "Source lines:",
        record[
            "source_line_count"
        ],
    )

    print(
        "Required conversion:",
        record[
            "conversion_type"
        ],
    )

    print()


print(
    "Missing target cells:",
    cell60_missing_target_numbers,
)

print(
    "Duplicate target cells:",
    cell60_duplicate_numbers,
)

print()

print("Checks:")

for (
    check_name,
    check_value,
) in cell60_checks.items():

    print(
        " -",
        check_name,
        ":",
        (
            "PASS"
            if check_value
            else "FAIL"
        ),
    )

print()

print(
    "CELL 60 STATUS:",
    cell60_status,
)

print(
    "Next readiness:",
    cell60_next_readiness,
)


assert cell60_status == "PASS", (
    "Cell 60 conversion-map audit failed."
)
````

<h2>مرحله ۶۱ — Final Run-All Health Check</h2>
<p>
پس از Restart Kernel → Run All، این مرحله وضعیت واقعی Notebook و Artifact نهایی Golden Dataset v2 را به‌صورت read-only بررسی می‌کند.

<p>
کنترل نهایی شامل اجرای همه Code Cellها، نبود خطای ذخیره‌شده، شماره‌گذاری کامل و یکتا، غیرفعال بودن تمام مسیرهای API، ثابت ماندن Dataset فریز‌شده، تعداد دقیق ۳۸۱۳ رکورد و Hash نهایی فایل Dataset است. این مرحله هیچ API call و هیچ File Write انجام نمی‌دهد.

````python
# Code Cell 61 — Final Run-All Health Check

from pathlib import Path
from collections import Counter
import hashlib
import json
import re


# =========================================================
# FROZEN EXPECTATIONS
# =========================================================

CELL61_EXPECTED_NOTEBOOK_NAME = (
    "12j_golden_dataset_v2_registry_driven_rebuild_RUN_ALL_SAFE.ipynb"
)

CELL61_EXPECTED_DATASET_RELATIVE_PATH = (
    Path("artifacts")
    / "phase2"
    / "golden_dataset_v2"
    / "golden_dataset_v2.jsonl"
)

CELL61_EXPECTED_RECORD_COUNT = 3813

CELL61_EXPECTED_DATASET_FILE_SHA256 = (
    "b9ca18e710686ddca48a25c637840b716adb306717cafd8ec7c96ce9e182e6f2"
)

# Cells 1–60 are the completed Run-All target.
# Cell 61 is the post-Run-All health checker itself.
CELL61_RUN_ALL_TARGET_NUMBERS = set(
    range(1, 61)
)

CELL61_HEALTH_CHECK_NUMBER = 61


# =========================================================
# ROOT + EXACT NOTEBOOK
# =========================================================

cell61_project_root = Path(
    project_root
).resolve()

cell61_notebook_candidates = sorted(
    cell61_project_root.rglob(
        CELL61_EXPECTED_NOTEBOOK_NAME
    )
)

assert len(cell61_notebook_candidates) == 1, (
    "Expected exactly one RUN_ALL_SAFE notebook. Found: "
    + repr(
        [
            str(
                path.relative_to(
                    cell61_project_root
                )
            )
            for path
            in cell61_notebook_candidates
        ]
    )
)

cell61_notebook_path = (
    cell61_notebook_candidates[0]
)


# =========================================================
# READ NOTEBOOK — NO WRITE
# =========================================================

with cell61_notebook_path.open(
    "r",
    encoding="utf-8",
) as handle:
    cell61_notebook = json.load(handle)

cell61_cells = cell61_notebook.get(
    "cells",
    [],
)


# =========================================================
# HELPERS
# =========================================================

def cell61_source_text(cell):

    source = cell.get(
        "source",
        [],
    )

    if isinstance(source, list):
        return "".join(source)

    if isinstance(source, str):
        return source

    return ""


def cell61_code_number(source_text):

    match = re.search(
        r"^\s*#\s*Code\s+Cell\s+(\d+)\b",
        source_text,
        flags=(
            re.IGNORECASE
            | re.MULTILINE
        ),
    )

    return (
        int(match.group(1))
        if match
        else None
    )


def cell61_sha256_file(path):

    digest = hashlib.sha256()

    with Path(path).open(
        "rb"
    ) as handle:

        for chunk in iter(
            lambda: handle.read(
                1024 * 1024
            ),
            b"",
        ):
            digest.update(chunk)

    return digest.hexdigest()


# =========================================================
# NOTEBOOK CODE-CELL INVENTORY
# =========================================================

cell61_code_records = []

for notebook_index, cell in enumerate(
    cell61_cells
):

    if (
        cell.get("cell_type")
        != "code"
    ):
        continue

    source_text = (
        cell61_source_text(
            cell
        )
    )

    number = cell61_code_number(
        source_text
    )

    outputs = cell.get(
        "outputs",
        [],
    )

    error_outputs = [
        output
        for output in outputs
        if (
            output.get(
                "output_type"
            )
            == "error"
        )
    ]

    cell61_code_records.append({
        "notebook_index": notebook_index,
        "code_number": number,
        "execution_count": cell.get(
            "execution_count"
        ),
        "error_outputs": error_outputs,
        "source_text": source_text,
    })


cell61_numbered_records = [
    record
    for record
    in cell61_code_records
    if (
        record[
            "code_number"
        ]
        is not None
    )
]


cell61_numbers = [
    record[
        "code_number"
    ]
    for record
    in cell61_numbered_records
]


cell61_number_counter = Counter(
    cell61_numbers
)


cell61_duplicate_numbers = sorted(
    number
    for number, count
    in cell61_number_counter.items()
    if count > 1
)


# =========================================================
# RUN-ALL TARGET = CELLS 1–60 ONLY
# =========================================================

cell61_run_all_target_records = [
    record
    for record
    in cell61_numbered_records
    if (
        record[
            "code_number"
        ]
        in CELL61_RUN_ALL_TARGET_NUMBERS
    )
]


cell61_run_all_target_numbers = {
    record[
        "code_number"
    ]
    for record
    in cell61_run_all_target_records
}


cell61_missing_run_all_numbers = sorted(
    CELL61_RUN_ALL_TARGET_NUMBERS
    - cell61_run_all_target_numbers
)


cell61_unexecuted_run_all_numbers = sorted(
    record[
        "code_number"
    ]
    for record
    in cell61_run_all_target_records
    if (
        record[
            "execution_count"
        ]
        is None
    )
)


cell61_run_all_error_numbers = sorted(
    record[
        "code_number"
    ]
    for record
    in cell61_run_all_target_records
    if record[
        "error_outputs"
    ]
)


# =========================================================
# HEALTH-CHECK CELL PRESENCE
# =========================================================

cell61_health_checker_records = [
    record
    for record
    in cell61_numbered_records
    if (
        record[
            "code_number"
        ]
        == CELL61_HEALTH_CHECK_NUMBER
    )
]


# =========================================================
# API SAFETY STATIC AUDIT
# ALL NUMBERED CELLS INCLUDING CELL 61
# =========================================================

cell61_api_true_assignments = []

for record in cell61_numbered_records:

    source_text = record[
        "source_text"
    ]

    true_matches = re.findall(
        r"(?m)^\s*"
        r"([A-Z][A-Z0-9_]*"
        r"(?:ALLOW_API|API_FOR_MISSING)"
        r"[A-Z0-9_]*)"
        r"\s*=\s*True\s*$",
        source_text,
    )

    for flag_name in true_matches:

        cell61_api_true_assignments.append({
            "code_cell": (
                record[
                    "code_number"
                ]
            ),
            "flag": flag_name,
        })


# =========================================================
# FROZEN DATASET HEALTH
# =========================================================

cell61_dataset_path = (
    cell61_project_root
    / CELL61_EXPECTED_DATASET_RELATIVE_PATH
)

cell61_dataset_exists = (
    cell61_dataset_path.is_file()
)

cell61_dataset_record_count = None
cell61_dataset_file_sha256 = None
cell61_dataset_parse_errors = []


if cell61_dataset_exists:

    cell61_dataset_file_sha256 = (
        cell61_sha256_file(
            cell61_dataset_path
        )
    )

    record_count = 0

    with cell61_dataset_path.open(
        "r",
        encoding="utf-8",
    ) as handle:

        for line_number, line in enumerate(
            handle,
            start=1,
        ):

            stripped = (
                line.strip()
            )

            if not stripped:
                continue

            try:
                json.loads(
                    stripped
                )

                record_count += 1

            except Exception as exc:

                cell61_dataset_parse_errors.append({
                    "line": line_number,
                    "error": repr(exc),
                })

    cell61_dataset_record_count = (
        record_count
    )


# =========================================================
# FINAL CHECKS
# =========================================================

cell61_checks = {

    "exact_run_all_safe_notebook_found": (
        len(
            cell61_notebook_candidates
        )
        == 1
    ),

    "run_all_target_cells_1_to_60_found": (
        cell61_run_all_target_numbers
        == CELL61_RUN_ALL_TARGET_NUMBERS
    ),

    "zero_missing_run_all_cells": (
        len(
            cell61_missing_run_all_numbers
        )
        == 0
    ),

    "health_check_cell_61_found_once": (
        len(
            cell61_health_checker_records
        )
        == 1
    ),

    "numbered_cells_exact_1_to_61": (
        sorted(
            cell61_numbers
        )
        == list(
            range(
                1,
                62,
            )
        )
    ),

    "zero_duplicate_code_numbers": (
        len(
            cell61_duplicate_numbers
        )
        == 0
    ),

    "all_run_all_cells_1_to_60_executed": (
        len(
            cell61_unexecuted_run_all_numbers
        )
        == 0
    ),

    "zero_run_all_error_outputs_1_to_60": (
        len(
            cell61_run_all_error_numbers
        )
        == 0
    ),

    "zero_api_enabled_true_flags": (
        len(
            cell61_api_true_assignments
        )
        == 0
    ),

    "dataset_exists": (
        cell61_dataset_exists
    ),

    "dataset_jsonl_parse_clean": (
        len(
            cell61_dataset_parse_errors
        )
        == 0
    ),

    "dataset_record_count_3813": (
        cell61_dataset_record_count
        == CELL61_EXPECTED_RECORD_COUNT
    ),

    "dataset_file_sha256_exact": (
        cell61_dataset_file_sha256
        == CELL61_EXPECTED_DATASET_FILE_SHA256
    ),

    "zero_api_calls_this_cell": True,

    "zero_file_writes_this_cell": True,
}


cell61_status = (
    "PASS"
    if all(
        cell61_checks.values()
    )
    else "FAIL"
)


cell61_readiness = (
    "NB12J_FINAL_RUN_ALL_HEALTH_PASS"
    if (
        cell61_status
        == "PASS"
    )
    else
    "NB12J_FINAL_RUN_ALL_HEALTH_REPAIR_REQUIRED"
)


# =========================================================
# OUTPUT
# =========================================================

print(
    "=== GOLDEN V2 FINAL RUN-ALL HEALTH CHECK ==="
)

print()

print(
    "Notebook:",
    str(
        cell61_notebook_path.relative_to(
            cell61_project_root
        )
    ),
)

print(
    "Notebook cells:",
    len(
        cell61_cells
    ),
)

print(
    "Numbered code cells:",
    len(
        cell61_numbered_records
    ),
)

print(
    "Run-All target:",
    "Code Cells 1–60",
)

print(
    "Missing Run-All cells:",
    cell61_missing_run_all_numbers,
)

print(
    "Unexecuted Run-All cells:",
    cell61_unexecuted_run_all_numbers,
)

print(
    "Run-All cells with saved errors:",
    cell61_run_all_error_numbers,
)

print(
    "Duplicate code numbers:",
    cell61_duplicate_numbers,
)

print(
    "API-enabled True assignments:",
    cell61_api_true_assignments,
)

print()

print(
    "Dataset:",
    str(
        CELL61_EXPECTED_DATASET_RELATIVE_PATH
    ),
)

print(
    "Dataset records:",
    cell61_dataset_record_count,
)

print(
    "Dataset file SHA256:",
    cell61_dataset_file_sha256,
)

print(
    "Expected file SHA256:",
    CELL61_EXPECTED_DATASET_FILE_SHA256,
)

print(
    "Dataset parse errors:",
    cell61_dataset_parse_errors,
)

print()

print(
    "Final checks:"
)

for (
    check_name,
    check_value,
) in cell61_checks.items():

    print(
        " -",
        check_name,
        ":",
        (
            "PASS"
            if check_value
            else "FAIL"
        ),
    )

print()

print(
    "CELL 61 STATUS:",
    cell61_status,
)

print(
    "Final readiness:",
    cell61_readiness,
)

assert (
    cell61_status
    == "PASS"
), (
    "NB12j final Run-All health check failed."
)
````
