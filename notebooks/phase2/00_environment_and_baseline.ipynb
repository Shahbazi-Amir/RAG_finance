```python
from pathlib import Path

current_working_directory = Path.cwd().resolve()
search_locations = [current_working_directory, *current_working_directory.parents]

project_root = None

for location in search_locations:
    required_paths = [
        location / ".git",
        location / "notebooks",
        location / "data",
    ]

    if all(path.exists() for path in required_paths):
        project_root = location
        break

print("Current working directory:", current_working_directory)

if project_root is None:
    raise RuntimeError(
        "Project root could not be identified from the current working directory."
    )

print("Project root:", project_root)
print("Git directory exists:", (project_root / ".git").exists())
print("Notebooks directory exists:", (project_root / "notebooks").exists())
print("Data directory exists:", (project_root / "data").exists())
```

### ثبت وضعیت Git به‌عنوان Baseline پروژه

**هدف:**  
ثبت وضعیت واقعی Repository در لحظه شروع رسمی فاز دوم.

**چرا این مرحله لازم است؟**  
برای اینکه نتایج Notebookهای بعدی قابل ردیابی و بازتولید باشند، باید مشخص باشد این اجرا دقیقاً روی کدام Branch و کدام Commit انجام شده است.

همچنین باید بدانیم Working Tree در زمان ثبت Baseline تمیز بوده یا فایل‌های تغییرکرده و Commit‌نشده در آن وجود داشته‌اند.

این اطلاعات بعداً در Dataset Release، Benchmarkها، گزارش‌های فنی و مستندات R&D به ما اجازه می‌دهد نتیجه هر اجرا را به نسخه دقیق کد متصل کنیم.

**کاری که انجام می‌شود:**  

- نام Branch فعال خوانده می‌شود.
- SHA کامل Commit جاری ثبت می‌شود.
- SHA کوتاه برای خوانایی نمایش داده می‌شود.
- آخرین Commit ثبت می‌شود.
- وضعیت Working Tree بررسی می‌شود.

این سلول فقط اطلاعات Git را می‌خواند و هیچ تغییری در Repository ایجاد نمی‌کند.

**معیار PASS:**  
Repository معتبر باشد و Branch و Commit جاری با موفقیت شناسایی شوند.

**معیار توجه:**  
اگر Working Tree دارای تغییرات Commit‌نشده باشد، اجرا متوقف نمی‌شود؛ اما این وضعیت باید به‌عنوان بخشی از Baseline ثبت و بعداً بررسی شود.

**خروجی مورد انتظار:**  
نمایش Branch، Commit SHA، آخرین Commit و وضعیت Clean یا Dirty بودن Repository.

</div>

```python
# Code Cell 03

import subprocess


def run_git_command(*args):
    result = subprocess.run(
        ["git", "-C", str(project_root), *args],
        capture_output=True,
        text=True,
        check=True,
    )
    return result.stdout.strip()


git_branch = run_git_command("branch", "--show-current")
git_commit_sha = run_git_command("rev-parse", "HEAD")
git_commit_short = run_git_command("rev-parse", "--short", "HEAD")
git_last_commit = run_git_command(
    "log",
    "-1",
    "--pretty=format:%h | %ad | %s",
    "--date=iso-strict",
)
git_status = run_git_command("status", "--porcelain")
git_working_tree = "clean" if not git_status else "dirty"

print("Git branch:", git_branch)
print("Git commit SHA:", git_commit_sha)
print("Git commit short:", git_commit_short)
print("Last commit:", git_last_commit)
print("Working tree:", git_working_tree)

if git_status:
    print("\nUncommitted changes:")
    print(git_status)
```

قبل از شروع پردازش این منابع باید مشخص شود در لحظه ثبت Baseline چه فایل‌ها و مجموعه‌هایی واقعاً در Repository وجود دارند.

این Inventory فقط وضعیت اولیه را ثبت می‌کند و هنوز درباره کیفیت، صحت، نسخه نهایی، حقوق محتوا یا قابل استفاده بودن فایل‌ها تصمیم‌گیری نمی‌کند.

**معیار PASS:**  
حداقل ساختار یکی از منابع قابل شناسایی باشد و عملیات Inventory بدون تغییر فایل‌ها انجام شود.

**نکته مهم:**  
تعداد فایل‌ها در این مرحله به معنی تعداد اسناد نهایی قابل استفاده در RAG نیست. تشخیص Duplicate، نسخه نهایی، Attribution، Rights و Quality در Notebookهای مخصوص هر منبع انجام خواهد شد.

</div>

```python
# Code Cell 05

from collections import OrderedDict

manifest_directory = project_root / "data" / "phase2" / "manifests"
books_directory = project_root / "data" / "books"
transcripts_final_directory = project_root / "data" / "transcripts" / "final"

website_manifests = sorted(
    path for path in manifest_directory.glob("*")
    if path.is_file()
)

book_files = sorted(
    path for path in books_directory.iterdir()
    if path.is_file()
)

transcript_collections = sorted(
    path for path in transcripts_final_directory.iterdir()
    if path.is_dir()
)

transcript_inventory = OrderedDict()

for collection in transcript_collections:
    markdown_files = sorted(collection.rglob("*.md"))
    transcript_inventory[collection.name] = len(markdown_files)

print("Website manifests:", len(website_manifests))
for path in website_manifests:
    print("  -", path.name)

print("\nBook files:", len(book_files))
for path in book_files:
    print("  -", path.name)

print("\nTranscript collections:", len(transcript_collections))
for collection_name, markdown_count in transcript_inventory.items():
    print(f"  - {collection_name}: {markdown_count} Markdown files")

print(
    "\nTotal Markdown files under transcripts/final:",
    sum(transcript_inventory.values()),
)
```

### ثبت هویت Manifest فعلی Website

**هدف:**  
ثبت مشخصات فنی فایل Manifest موجود برای Website به‌عنوان بخشی از Baseline فاز دوم.

**چرا این مرحله لازم است؟**  
خروجی Website قبلاً تولید شده است و قرار نیست در این Notebook دوباره Crawl یا پردازش شود.

قبل از اینکه در Notebook مخصوص Website Acceptance کیفیت این خروجی بررسی شود، باید دقیقاً مشخص کنیم Acceptance روی کدام فایل و کدام نسخه از Manifest انجام خواهد شد.

برای این منظور نام فایل، اندازه، Hash و ساختار سطح بالای JSON ثبت می‌شود.

Hash فایل به ما اجازه می‌دهد بعداً تشخیص دهیم آیا Manifest از زمان ثبت Baseline تغییر کرده است یا خیر.

**کاری که انجام می‌شود:**  

- فایل `website_manifest_v1.json` خوانده می‌شود.
- SHA-256 آن محاسبه می‌شود.
- اندازه فایل ثبت می‌شود.
- نوع ساختار اصلی JSON مشخص می‌شود.
- در صورت امکان تعداد رکوردهای سطح اصلی نمایش داده می‌شود.
- اگر ساختار JSON دیکشنری باشد، کلیدهای سطح اول نمایش داده می‌شوند.

در این مرحله هیچ Quality Check مربوط به محتوای مقالات انجام نمی‌شود.

**معیار PASS:**  
Manifest موجود باشد، JSON با موفقیت خوانده شود و SHA-256 آن قابل محاسبه باشد.

**معیار FAIL:**  
فایل وجود نداشته باشد، JSON قابل خواندن نباشد یا فایل در زمان خواندن دچار خطا شود.

</div>

```python
# Code Cell 06

import hashlib
import json

website_manifest_path = (
    project_root
    / "data"
    / "phase2"
    / "manifests"
    / "website_manifest_v1.json"
)

assert website_manifest_path.exists(), (
    f"Website manifest not found: {website_manifest_path}"
)

manifest_bytes = website_manifest_path.read_bytes()
manifest_sha256 = hashlib.sha256(manifest_bytes).hexdigest()

with website_manifest_path.open("r", encoding="utf-8") as file:
    website_manifest = json.load(file)

print("Manifest file:", website_manifest_path.name)
print("Manifest size (bytes):", website_manifest_path.stat().st_size)
print("Manifest SHA-256:", manifest_sha256)
print("Top-level JSON type:", type(website_manifest).__name__)

if isinstance(website_manifest, list):
    print("Top-level record count:", len(website_manifest))

elif isinstance(website_manifest, dict):
    print("Top-level keys:", list(website_manifest.keys()))

    for key, value in website_manifest.items():
        if isinstance(value, list):
            print(f"List field '{key}' contains:", len(value), "items")

print("\nWebsite manifest baseline gate: PASS")
```

### تفسیر Baseline مربوط به Website

```python
# Code Cell 07

dependency_candidates = [
    "pyproject.toml",
    "requirements.txt",
    "requirements-dev.txt",
    "environment.yml",
    "environment.yaml",
    "Pipfile",
    "Pipfile.lock",
    "poetry.lock",
    "setup.py",
    "setup.cfg",
]

dependency_files = []

for filename in dependency_candidates:
    path = project_root / filename
    if path.is_file():
        dependency_files.append(path)

for path in sorted(project_root.glob("requirements*.txt")):
    if path.is_file() and path not in dependency_files:
        dependency_files.append(path)

dependency_files = sorted(dependency_files, key=lambda path: path.name.lower())

print("Dependency/environment files found:", len(dependency_files))

for path in dependency_files:
    print(
        f"  - {path.name} "
        f"({path.stat().st_size} bytes)"
    )

if not dependency_files:
    print("  - No recognized dependency/environment file found at repository root.")

print("\nDependency declaration discovery gate: PASS")
```

### ثبت نسخه واقعی Dependencyهای نصب‌شده

**هدف:**  
ثبت نسخه Packageهایی که پروژه در `requirements.txt` تعریف کرده و اکنون در Environment فعال نصب هستند.

**چرا این مرحله لازم است؟**  
چون `requirements.txt` فعلی Version Pin ندارد، فقط از روی آن نمی‌توان فهمید Notebook در زمان ثبت Baseline با چه نسخه‌هایی اجرا شده است.

ثبت نسخه واقعی Packageهای نصب‌شده باعث می‌شود بعداً بتوانیم تغییرات Environment را تشخیص دهیم و در صورت تفاوت نتیجه‌ها، یکی از عوامل احتمالی را بررسی کنیم.

**کاری که انجام می‌شود:**  

برای هر یک از Dependencyهای اعلام‌شده:

- وضعیت نصب بررسی می‌شود.
- نسخه نصب‌شده خوانده می‌شود.
- اگر Package موجود نباشد، وضعیت آن به‌صورت `NOT INSTALLED` ثبت می‌شود.

هیچ Packageای در این مرحله نصب یا Upgrade نمی‌شود.

**معیار PASS:**  
تمام Dependencyهای تعریف‌شده بررسی شوند.

وجود Package نصب‌نشده مخفی نمی‌شود و در خروجی به‌وضوح نمایش داده خواهد شد.

</div>

```python
# Code Cell 09

from importlib.metadata import PackageNotFoundError, version

declared_packages = {
    "streamlit": "streamlit",
    "python-dotenv": "python-dotenv",
    "openai": "openai",
    "qdrant-client": "qdrant-client",
    "requests": "requests",
    "beautifulsoup4": "beautifulsoup4",
    "lxml": "lxml",
    "tqdm": "tqdm",
}

installed_dependency_versions = {}

for requirement_name, package_name in declared_packages.items():
    try:
        installed_version = version(package_name)
        installed_dependency_versions[requirement_name] = installed_version
        print(f"{requirement_name:<20} {installed_version}")
    except PackageNotFoundError:
        installed_dependency_versions[requirement_name] = None
        print(f"{requirement_name:<20} NOT INSTALLED")

missing_dependencies = [
    name
    for name, installed_version in installed_dependency_versions.items()
    if installed_version is None
]

print("\nDeclared dependencies:", len(declared_packages))
print("Installed dependencies:", len(declared_packages) - len(missing_dependencies))
print("Missing dependencies:", len(missing_dependencies))

print("\nDependency environment inspection gate: PASS")
```
