```python
# Code Cell 01 — Centralized imports / bootstrap
from __future__ import annotations

import hashlib
import json
import re
from pathlib import Path
from typing import Any

NOTEBOOK_NAME = "12_golden_dataset.ipynb"
EXPECTED_RECORD_COUNT = 3813
EXPECTED_DATASET_SHA256 = "b9ca18e710686ddca48a25c637840b716adb306717cafd8ec7c96ce9e182e6f2"
EXPECTED_PAYLOAD_SHA256 = "539635b4c89e8f4eda41dde8106e04307c063f3fe0f5db402ef0440802233a03"
READ_ONLY_NOTEBOOK = True
```

```python
# Code Cell 02 — Repository discovery and read-only preflight
def find_repo_root() -> Path:
    start = Path.cwd().resolve()
    for candidate in [start, *start.parents]:
        if (candidate / "notebooks" / "phase2").is_dir() and (candidate / "artifacts" / "phase2").is_dir():
            return candidate
    raise FileNotFoundError("Repository root with notebooks/phase2 and artifacts/phase2 was not found.")

REPO_ROOT = find_repo_root()
assert READ_ONLY_NOTEBOOK is True
print("Repository root:", REPO_ROOT)
print("Execution policy: READ ONLY")
print("PRE-FLIGHT: PASS")
```

<div style="direction: rtl; white-space: normal; line-height: 1.8;">
<h2>Artifactهای قطعی و فریز‌شده</h2><p>مسیرهای زیر، دادهٔ معتبر نهایی هستند. هیچ Cellی در این نوت‌بوک مجاز به نوشتن روی آن‌ها نیست.</p>

```python
# Code Cell 03 — Register canonical artifact paths
PATHS = {
    "dataset": REPO_ROOT / "artifacts/phase2/golden_dataset_v2/golden_dataset_v2.jsonl",
    "manifest": REPO_ROOT / "artifacts/phase2/golden_dataset_v2/golden_dataset_v2_manifest.json",
    "final_report": REPO_ROOT / "artifacts/phase2/golden_dataset_v2/golden_dataset_v2_final_report.json",
    "registry_report": REPO_ROOT / "artifacts/phase2/knowledge_registry/claim_registry_report_v1.json",
    "archive": REPO_ROOT / "notebooks/phase2/archive/series12_notebooks",
    "roadmap": REPO_ROOT / "data/Road_Maps/RAG_finance_phase2_modular_notebooks_execution_plan.md",
}
missing = [name for name, path in PATHS.items() if name != "archive" and not path.is_file()]
assert not missing, f"Required artifact(s) missing: {missing}"
print("Required canonical artifacts:", len(PATHS) - 1)
print("ARTIFACT PATH PREFLIGHT: PASS")
```

```python
# Code Cell 04 — Read frozen manifest and final report
golden_manifest = json.loads(PATHS["manifest"].read_text(encoding="utf-8"))
golden_final_report = json.loads(PATHS["final_report"].read_text(encoding="utf-8"))

manifest_checks = {
    "manifest_frozen": golden_manifest.get("freeze_status") == "FROZEN",
    "manifest_record_count": golden_manifest.get("record_count") == EXPECTED_RECORD_COUNT,
    "manifest_semantic_pass": golden_manifest.get("semantic_distribution", {}).get("PASS") == EXPECTED_RECORD_COUNT,
    "manifest_semantic_fail_zero": golden_manifest.get("semantic_distribution", {}).get("FAIL") == 0,
    "final_report_pass": golden_final_report.get("status") == "PASS",
    "final_report_frozen": golden_final_report.get("freeze_status") == "FROZEN",
}
for name, passed in manifest_checks.items():
    print(f"{name}: {'PASS' if passed else 'FAIL'}")
assert all(manifest_checks.values()), "Frozen manifest/final-report validation failed."
print("FROZEN MANIFEST AND FINAL REPORT: PASS")
```

```python
# Code Cell 05 — Validate frozen JSONL, count and file hash
dataset_sha256 = hashlib.sha256()
records: list[dict[str, Any]] = []
parse_errors: list[str] = []

with PATHS["dataset"].open("r", encoding="utf-8") as handle:
    for line_number, line in enumerate(handle, start=1):
        dataset_sha256.update(line.encode("utf-8"))
        if not line.strip():
            continue
        try:
            records.append(json.loads(line))
        except json.JSONDecodeError as exc:
            parse_errors.append(f"line {line_number}: {exc}")

observed_dataset_sha256 = dataset_sha256.hexdigest()
assert not parse_errors, f"JSONL parse errors: {parse_errors[:3]}"
assert len(records) == EXPECTED_RECORD_COUNT, f"Expected {EXPECTED_RECORD_COUNT}, found {len(records)}."
assert observed_dataset_sha256 == EXPECTED_DATASET_SHA256, "Frozen dataset SHA256 mismatch."

print("JSONL records:", len(records))
print("JSONL parse errors:", len(parse_errors))
print("Dataset SHA256:", observed_dataset_sha256)
print("FROZEN DATASET INTEGRITY: PASS")
```

```python
# Code Cell 06 — Validate record identity and semantic acceptance
required_fields = {
    "candidate_id", "claim_id", "claim_text_fa", "evaluation_content",
    "evidence_ids", "source_ids", "generation", "validation",
}
missing_field_records = []
candidate_ids, claim_ids = set(), set()
semantic_failures = []

for index, record in enumerate(records, start=1):
    missing_fields = sorted(required_fields - set(record))
    if missing_fields:
        missing_field_records.append((index, missing_fields))
    candidate_ids.add(record.get("candidate_id"))
    claim_ids.add(record.get("claim_id"))
    if record.get("validation", {}).get("semantic_verdict") != "PASS":
        semantic_failures.append(record.get("candidate_id"))

assert not missing_field_records, f"Records with missing required fields: {missing_field_records[:3]}"
assert len(candidate_ids) == EXPECTED_RECORD_COUNT, "candidate_id values are not unique."
assert len(claim_ids) == EXPECTED_RECORD_COUNT, "claim_id values are not unique."
assert not semantic_failures, f"Semantic failures found: {semantic_failures[:3]}"

print("Unique candidate IDs:", len(candidate_ids))
print("Unique claim IDs:", len(claim_ids))
print("Semantic PASS:", len(records))
print("RECORD IDENTITY AND SEMANTIC VALIDATION: PASS")
```

```python
# Code Cell 12 — Final notebook health check
def locate_this_notebook() -> Path:
    candidates = [
        REPO_ROOT / "notebooks/phase2/12_golden_dataset.ipynb",
        Path.cwd() / "12_golden_dataset.ipynb",
    ]

    for candidate in candidates:
        if candidate.is_file():
            return candidate

    raise FileNotFoundError(
        "Save this notebook as notebooks/phase2/12_golden_dataset.ipynb "
        "before final health check."
    )


notebook_path = locate_this_notebook()
notebook_json = json.loads(
    notebook_path.read_text(encoding="utf-8")
)

code_cells = [
    cell
    for cell in notebook_json.get("cells", [])
    if cell.get("cell_type") == "code"
]

number_pattern = re.compile(
    r"(?m)^\s*#\s*Code Cell\s+(\d+)\b"
)

numbers = []
errors = []
unexecuted = []

for cell in code_cells:
    source_value = cell.get("source", [])

    source = (
        "".join(source_value)
        if isinstance(source_value, list)
        else str(source_value)
    )

    match = number_pattern.search(source)

    if match:
        numbers.append(int(match.group(1)))

    if cell.get("execution_count") is None:
        unexecuted.append(
            match.group(1) if match else "unnumbered"
        )

    errors.extend(
        output
        for output in cell.get("outputs", [])
        if output.get("output_type") == "error"
    )

first_source_value = (
    code_cells[0].get("source", [])
    if code_cells
    else []
)

first_code_source = (
    "".join(first_source_value)
    if isinstance(first_source_value, list)
    else str(first_source_value)
)

expected_numbers = list(
    range(1, len(code_cells) + 1)
)

final_cell_number = str(len(code_cells))

unexecuted_required = [
    number
    for number in unexecuted
    if number != final_cell_number
]

health_checks = {
    "sequential_code_numbers": (
        numbers == expected_numbers
    ),
    "no_unexecuted_code_cells": (
        not unexecuted_required
    ),
    "no_saved_error_outputs": (
        not errors
    ),
    "bootstrap_is_first_cell": bool(
        re.search(
            r"(?m)^\s*#\s*Code Cell\s+01\b",
            first_code_source,
        )
    ),
}

for name, passed in health_checks.items():
    print(
        f"{name}: {'PASS' if passed else 'FAIL'}"
    )

print(
    "NOTEBOOK HEALTH:",
    "PASS" if all(health_checks.values()) else "FAIL",
)

assert all(health_checks.values()), (
    "Clear any previous error output, save the notebook, "
    "then run this health-check cell again."
)
```
