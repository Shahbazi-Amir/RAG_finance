```python
# Code Cell 03

corpus_path = required_paths["accepted_corpus"]

root_cause = parsed_json["root_cause_analysis"]
structural_benchmark = parsed_json["structural_benchmark"]
handoff = parsed_json["nb12_handoff"]

def percentile(values, q):
    if not values:
        return None

    ordered = sorted(values)

    if len(ordered) == 1:
        return ordered[0]

    position = (len(ordered) - 1) * q
    lower = int(position)
    upper = min(lower + 1, len(ordered) - 1)
    fraction = position - lower

    return round(
        ordered[lower] * (1 - fraction)
        + ordered[upper] * fraction,
        2
    )


def scalarize(value):
    if isinstance(value, (str, int, float, bool)) or value is None:
        return str(value)

    try:
        return json.dumps(
            value,
            ensure_ascii=False,
            sort_keys=True
        )
    except Exception:
        return repr(value)


def collect_nested_scalar_paths(
    obj,
    prefix="",
    selected_terms=None,
    output=None,
):
    if output is None:
        output = []

    if selected_terms is None:
        selected_terms = set()

    if isinstance(obj, dict):
        for key, value in obj.items():
            path = f"{prefix}.{key}" if prefix else str(key)

            if isinstance(value, (dict, list)):
                collect_nested_scalar_paths(
                    value,
                    prefix=path,
                    selected_terms=selected_terms,
                    output=output,
                )
            else:
                lowered = path.lower()

                if any(term in lowered for term in selected_terms):
                    output.append((path, scalarize(value)))

    elif isinstance(obj, list):
        for index, value in enumerate(obj):
            path = f"{prefix}[]"

            if isinstance(value, (dict, list)):
                collect_nested_scalar_paths(
                    value,
                    prefix=path,
                    selected_terms=selected_terms,
                    output=output,
                )

    return output


website_lengths = []
book_lengths = []
transcript_lengths = []

website_origin_lengths = defaultdict(list)

book_unit_type = Counter()
book_in_table = Counter()
book_is_heading = Counter()
book_heading_level = Counter()
book_heading_depth = Counter()
book_content_type = Counter()

transcript_program_id = Counter()
transcript_program_name = Counter()
transcript_physical_file_id = Counter()
transcript_episode_number = Counter()
transcript_season_number = Counter()

transcript_attribution_speaker = Counter()
transcript_metadata_speaker_type = Counter()
transcript_metadata_speaker_value = Counter()

transcript_record_type = Counter()
transcript_processing_state = Counter()

nested_signal_paths = {
    "website": defaultdict(Counter),
    "book": defaultdict(Counter),
    "transcript": defaultdict(Counter),
}

selected_signal_terms = {
    "primary",
    "speaker",
    "authority",
    "role",
    "program",
    "season",
    "episode",
    "physical_file",
    "utterance",
    "table",
    "heading",
    "content_type",
    "record_type",
    "parent",
    "supplement",
}

source_samples = {}

with corpus_path.open("r", encoding="utf-8") as f:
    for line in f:
        row = json.loads(line)

        source_type = row["source_type"]
        text = row.get("text") or ""
        text_length = len(text)

        attribution = row.get("attribution") or {}
        metadata = row.get("source_metadata") or {}
        locator = row.get("locator") or {}

        if source_type not in source_samples:
            source_samples[source_type] = {
                "document_id": row.get("document_id"),
                "attribution": attribution,
                "source_metadata": metadata,
                "locator": locator,
                "content_origin": row.get("content_origin"),
            }

        if source_type == "website":
            website_lengths.append(text_length)

            origin = str(row.get("content_origin"))
            website_origin_lengths[origin].append(text_length)

        elif source_type == "book":
            book_lengths.append(text_length)

            book_unit_type[
                scalarize(metadata.get("book_unit_type"))
            ] += 1

            book_in_table[
                scalarize(metadata.get("in_table"))
            ] += 1

            book_is_heading[
                scalarize(metadata.get("is_heading"))
            ] += 1

            book_heading_level[
                scalarize(metadata.get("heading_level"))
            ] += 1

            heading_path = (
                locator.get("heading_path")
                or metadata.get("heading_path")
            )

            if isinstance(heading_path, list):
                depth = len(heading_path)
            elif isinstance(heading_path, str) and heading_path.strip():
                depth = len(
                    [
                        part
                        for part in re.split(
                            r"\s*(?:>|/|→|\||»)\s*",
                            heading_path
                        )
                        if part.strip()
                    ]
                )
            else:
                depth = 0

            book_heading_depth[depth] += 1

            book_content_type[
                scalarize(metadata.get("content_type"))
            ] += 1

        elif source_type == "transcript":
            transcript_lengths.append(text_length)

            transcript_program_id[
                scalarize(locator.get("program_id"))
            ] += 1

            transcript_program_name[
                scalarize(metadata.get("program"))
            ] += 1

            transcript_physical_file_id[
                scalarize(locator.get("physical_file_id"))
            ] += 1

            transcript_episode_number[
                scalarize(locator.get("episode_number"))
            ] += 1

            transcript_season_number[
                scalarize(locator.get("season_number"))
            ] += 1

            transcript_attribution_speaker[
                scalarize(attribution.get("speaker"))
            ] += 1

            metadata_speaker = metadata.get("speaker")

            transcript_metadata_speaker_type[
                type(metadata_speaker).__name__
            ] += 1

            transcript_metadata_speaker_value[
                scalarize(metadata_speaker)
            ] += 1

            transcript_record_type[
                scalarize(metadata.get("record_type"))
            ] += 1

            transcript_processing_state[
                scalarize(metadata.get("processing_state"))
            ] += 1

        nested_pairs = collect_nested_scalar_paths(
            {
                "attribution": attribution,
                "source_metadata": metadata,
                "locator": locator,
            },
            selected_terms=selected_signal_terms,
        )

        for path, value in nested_pairs:
            nested_signal_paths[source_type][path][value] += 1


print("=== REQUIRED NB12 STRATA ===")

for source_group in ["website", "book", "transcript"]:
    print(f"\n[{source_group.upper()}]")
    for stratum in handoff[source_group]:
        print("-", stratum)


print("\n\n=== WEBSITE STRUCTURAL SIGNALS ===")

print("records:", len(website_lengths))
print(
    "text_length_chars:",
    {
        "min": min(website_lengths),
        "p25": percentile(website_lengths, 0.25),
        "p50": percentile(website_lengths, 0.50),
        "p75": percentile(website_lengths, 0.75),
        "p90": percentile(website_lengths, 0.90),
        "max": max(website_lengths),
    }
)

print("\nLength by content_origin:")

for origin, values in sorted(website_origin_lengths.items()):
    print(
        origin,
        {
            "count": len(values),
            "min": min(values),
            "p50": percentile(values, 0.50),
            "p90": percentile(values, 0.90),
            "max": max(values),
        }
    )


print("\n\n=== BOOK STRUCTURAL SIGNALS ===")

print("records:", len(book_lengths))
print(
    "text_length_chars:",
    {
        "min": min(book_lengths),
        "p25": percentile(book_lengths, 0.25),
        "p50": percentile(book_lengths, 0.50),
        "p75": percentile(book_lengths, 0.75),
        "p90": percentile(book_lengths, 0.90),
        "max": max(book_lengths),
    }
)

print("\nbook_unit_type:")
print(book_unit_type)

print("\nin_table:")
print(book_in_table)

print("\nis_heading:")
print(book_is_heading)

print("\nheading_level:")
print(book_heading_level)

print("\nheading_depth:")
print(sorted(book_heading_depth.items()))

print("\ncontent_type:")
print(book_content_type)


print("\n\n=== TRANSCRIPT STRUCTURAL SIGNALS ===")

print("records:", len(transcript_lengths))
print(
    "text_length_chars:",
    {
        "min": min(transcript_lengths),
        "p25": percentile(transcript_lengths, 0.25),
        "p50": percentile(transcript_lengths, 0.50),
        "p75": percentile(transcript_lengths, 0.75),
        "p90": percentile(transcript_lengths, 0.90),
        "max": max(transcript_lengths),
    }
)

print("\nprogram_id:")
for key, value in transcript_program_id.most_common():
    print("-", key, ":", value)

print("\nprogram:")
for key, value in transcript_program_name.most_common():
    print("-", key, ":", value)

print(
    "\nunique physical_file_id:",
    len(transcript_physical_file_id)
)

print("\nepisode_number:")
print(transcript_episode_number)

print("\nseason_number:")
print(transcript_season_number)

print("\nattribution.speaker:")
print(transcript_attribution_speaker)

print("\nsource_metadata.speaker types:")
print(transcript_metadata_speaker_type)

print("\nsource_metadata.speaker values — top 25:")
for key, value in transcript_metadata_speaker_value.most_common(25):
    print("-", key, ":", value)

print("\nrecord_type:")
print(transcript_record_type)

print("\nprocessing_state:")
print(transcript_processing_state)


print("\n\n=== SELECTED NESTED SIGNAL PATHS ===")

for source_type in ["website", "book", "transcript"]:
    print(f"\n[{source_type.upper()}]")

    for path in sorted(nested_signal_paths[source_type]):
        counts = nested_signal_paths[source_type][path]

        print(
            path,
            "->",
            counts.most_common(12)
        )


print("\n\n=== REPRESENTATIVE NON-TEXT SCHEMA SAMPLES ===")

for source_type in ["website", "book", "transcript"]:
    print(f"\n[{source_type.upper()}]")

    sample = source_samples[source_type]

    print(
        json.dumps(
            sample,
            ensure_ascii=False,
            indent=2
        )
    )


print("\n\n=== NB11 ROOT-CAUSE / STRUCTURAL REPORT KEYS ===")

print(
    "root_cause_analysis keys:",
    sorted(root_cause.keys())
)

print(
    "structural_benchmark keys:",
    sorted(structural_benchmark.keys())
)


print("\nCODE CELL 03 STATUS: INSPECTION_COMPLETE")
```

اگر Metadata مستقیم برای چنین رابطه‌ای وجود نداشته باشد، آن محدودیت باید صریحاً ثبت شود.

<li>تشخیص اینکه برای تعریف Contract چه مواردی Metadata-driven و چه مواردی نیازمند Review یا Token Measurement هستند.</li>
</ul>

```python
# Code Cell 04

def recursive_matches(obj, search_terms, prefix="", matches=None):
    if matches is None:
        matches = []

    if isinstance(obj, dict):
        for key, value in obj.items():
            path = f"{prefix}.{key}" if prefix else str(key)
            lowered_path = path.lower()

            if any(term in lowered_path for term in search_terms):
                if not isinstance(value, (dict, list)):
                    matches.append(
                        (path, scalarize(value))
                    )

            recursive_matches(
                value,
                search_terms,
                prefix=path,
                matches=matches,
            )

    elif isinstance(obj, list):
        for index, value in enumerate(obj):
            path = f"{prefix}[{index}]"

            recursive_matches(
                value,
                search_terms,
                prefix=path,
                matches=matches,
            )

    return matches


transcript_files = defaultdict(list)

with corpus_path.open("r", encoding="utf-8") as f:
    for line in f:
        row = json.loads(line)

        if row.get("source_type") != "transcript":
            continue

        metadata = row.get("source_metadata") or {}
        locator = row.get("locator") or {}
        speaker = metadata.get("speaker") or {}

        base = speaker.get("base") or {}
        effective = speaker.get("effective") or {}

        physical_file_id = locator.get("physical_file_id")
        utterance_index = metadata.get("utterance_index")

        transcript_files[physical_file_id].append(
            {
                "document_id": row.get("document_id"),
                "program_id": locator.get("program_id"),
                "program": metadata.get("program"),
                "episode_number": locator.get("episode_number"),
                "season_number": locator.get("season_number"),
                "utterance_index": utterance_index,
                "effective_label": effective.get("label"),
                "effective_canonical": effective.get("canonical"),
                "primary_dr_roudi_corpus": effective.get(
                    "primary_dr_roudi_corpus"
                ),
                "base_status": base.get("status"),
                "base_raw": base.get("raw"),
                "boundary_type": base.get("boundary_type"),
                "text": row.get("text") or "",
            }
        )


for physical_file_id in transcript_files:
    transcript_files[physical_file_id].sort(
        key=lambda x: (
            x["utterance_index"] is None,
            x["utterance_index"]
            if x["utterance_index"] is not None
            else 10**12,
            x["document_id"] or "",
        )
    )


file_profiles = []

transition_counts = Counter()
transition_program_counts = defaultdict(Counter)

for physical_file_id, rows in transcript_files.items():
    labels = {
        row["effective_label"]
        for row in rows
        if row["effective_label"] is not None
    }

    canonicals = {
        row["effective_canonical"]
        for row in rows
        if row["effective_canonical"] is not None
    }

    raw_labels = {
        row["base_raw"]
        for row in rows
        if row["base_raw"] is not None
    }

    primary_states = Counter(
        scalarize(row["primary_dr_roudi_corpus"])
        for row in rows
    )

    unresolved_rows = sum(
        1
        for row in rows
        if row["base_status"] == "UNRESOLVED"
    )

    file_profiles.append(
        {
            "physical_file_id": physical_file_id,
            "program_id": rows[0]["program_id"],
            "program": rows[0]["program"],
            "episode_number": rows[0]["episode_number"],
            "season_number": rows[0]["season_number"],
            "rows": len(rows),
            "labels": labels,
            "canonicals": canonicals,
            "raw_labels": raw_labels,
            "primary_states": primary_states,
            "unresolved_rows": unresolved_rows,
        }
    )

    for left, right in zip(rows, rows[1:]):
        left_state = left["primary_dr_roudi_corpus"]
        right_state = right["primary_dr_roudi_corpus"]

        if left_state is True:
            left_class = "PRIMARY"
        elif left_state is False:
            left_class = "NONPRIMARY_CONFIRMED"
        else:
            left_class = "UNRESOLVED_OR_CONTEXT"

        if right_state is True:
            right_class = "PRIMARY"
        elif right_state is False:
            right_class = "NONPRIMARY_CONFIRMED"
        else:
            right_class = "UNRESOLVED_OR_CONTEXT"

        transition = f"{left_class} -> {right_class}"

        transition_counts[transition] += 1
        transition_program_counts[
            left["program_id"]
        ][transition] += 1


single_label_files = [
    profile
    for profile in file_profiles
    if len(profile["labels"]) == 1
]

multi_label_files = [
    profile
    for profile in file_profiles
    if len(profile["labels"]) > 1
]

all_primary_files = [
    profile
    for profile in file_profiles
    if profile["primary_states"]
    and set(profile["primary_states"]) == {"True"}
]

all_unresolved_files = [
    profile
    for profile in file_profiles
    if profile["rows"] > 0
    and profile["unresolved_rows"] == profile["rows"]
]

mixed_primary_signal_files = [
    profile
    for profile in file_profiles
    if len(profile["primary_states"]) > 1
]


required_multi_speaker_programs = {
    "program:uni-tehran",
    "program:chehelston",
    "program:asre-shirin",
    "program:khane-to",
    "program:mizan",
}

program_file_profiles = defaultdict(list)

for profile in file_profiles:
    program_file_profiles[
        profile["program_id"]
    ].append(profile)


print("=== TRANSCRIPT FILE-LEVEL PROFILE ===")
print("physical files:", len(file_profiles))
print("single effective-label files:", len(single_label_files))
print("multi effective-label files:", len(multi_label_files))
print("all-primary files:", len(all_primary_files))
print("all-unresolved files:", len(all_unresolved_files))
print(
    "mixed primary-signal files:",
    len(mixed_primary_signal_files)
)


print("\n=== REQUIRED MULTI-SPEAKER PROGRAMS ===")

for program_id in sorted(required_multi_speaker_programs):
    profiles = program_file_profiles.get(program_id, [])

    multi = [
        p
        for p in profiles
        if len(p["labels"]) > 1
    ]

    unresolved = [
        p
        for p in profiles
        if p["unresolved_rows"] > 0
    ]

    mixed_primary = [
        p
        for p in profiles
        if len(p["primary_states"]) > 1
    ]

    total_rows = sum(
        p["rows"]
        for p in profiles
    )

    print(
        program_id,
        {
            "files": len(profiles),
            "rows": total_rows,
            "multi_effective_label_files": len(multi),
            "files_with_unresolved_rows": len(unresolved),
            "mixed_primary_signal_files": len(mixed_primary),
        }
    )


print("\n=== ADJACENT PRIMARY-STATE TRANSITIONS ===")

for transition, count in transition_counts.most_common():
    print("-", transition, ":", count)


print(
    "\nPRIMARY -> UNRESOLVED_OR_CONTEXT by required program:"
)

for program_id in sorted(required_multi_speaker_programs):
    count = transition_program_counts[
        program_id
    ]["PRIMARY -> UNRESOLVED_OR_CONTEXT"]

    print("-", program_id, ":", count)


print(
    "\nUNRESOLVED_OR_CONTEXT -> PRIMARY by required program:"
)

for program_id in sorted(required_multi_speaker_programs):
    count = transition_program_counts[
        program_id
    ]["UNRESOLVED_OR_CONTEXT -> PRIMARY"]

    print("-", program_id, ":", count)


print("\n=== REPRESENTATIVE FILE PROFILES ===")

representative_groups = {
    "all_primary": all_primary_files[:3],
    "all_unresolved": all_unresolved_files[:3],
    "multi_label": multi_label_files[:5],
    "mixed_primary_signal": mixed_primary_signal_files[:5],
}

for group_name, profiles in representative_groups.items():
    print(f"\n[{group_name}]")

    for profile in profiles:
        print(
            {
                "program_id": profile["program_id"],
                "episode_number": profile["episode_number"],
                "season_number": profile["season_number"],
                "rows": profile["rows"],
                "labels": sorted(
                    scalarize(x)
                    for x in profile["labels"]
                ),
                "raw_labels": sorted(
                    scalarize(x)
                    for x in profile["raw_labels"]
                ),
                "primary_states": dict(
                    profile["primary_states"]
                ),
                "unresolved_rows": profile["unresolved_rows"],
            }
        )


print("\n=== QUESTION / ANSWER DIRECT METADATA SIGNAL ===")

qa_paths = set()

for source_type, path_map in nested_signal_paths.items():
    for path in path_map:
        lowered = path.lower()

        if (
            "question" in lowered
            or "answer" in lowered
            or "qa" in lowered
        ):
            qa_paths.add(
                (source_type, path)
            )

if qa_paths:
    for source_type, path in sorted(qa_paths):
        print("-", source_type, ":", path)
else:
    print(
        "NO_DIRECT_QA_METADATA_SIGNAL_FOUND"
    )


print("\n=== NB11 TOKEN / OVERSIZE / FALLBACK EVIDENCE ===")

search_terms = {
    "oversize",
    "fallback",
    "sentence",
    "500",
    "800",
    "utterance",
    "token",
}

report_sources = {
    "root_cause_analysis": root_cause,
    "structural_benchmark": structural_benchmark,
    "chunk_contract": parsed_json["chunk_contract"],
}

for report_name, report in report_sources.items():
    print(f"\n[{report_name}]")

    matches = recursive_matches(
        report,
        search_terms,
    )

    unique_matches = []

    seen = set()

    for path, value in matches:
        key = (path, value)

        if key in seen:
            continue

        seen.add(key)
        unique_matches.append(
            (path, value)
        )

    if unique_matches:
        for path, value in unique_matches:
            print("-", path, "=", value)
    else:
        print("NO_MATCHING_SIGNAL")


print("\n=== INSPECTION INTERPRETATION FLAGS ===")

inspection_flags = {
    "required_multispeaker_programs_present":
        all(
            program_file_profiles.get(program_id)
            for program_id
            in required_multi_speaker_programs
        ),

    "primary_transition_candidates_present":
        transition_counts[
            "PRIMARY -> UNRESOLVED_OR_CONTEXT"
        ] > 0,

    "reverse_transition_candidates_present":
        transition_counts[
            "UNRESOLVED_OR_CONTEXT -> PRIMARY"
        ] > 0,

    "mixed_primary_signal_files_present":
        len(mixed_primary_signal_files) > 0,

    "single_primary_file_candidates_present":
        len(all_primary_files) > 0,

    "single_unresolved_file_candidates_present":
        len(all_unresolved_files) > 0,

    "direct_qa_metadata_present":
        bool(qa_paths),
}

for name, value in inspection_flags.items():
    print(
        name,
        ":",
        "YES" if value else "NO"
    )


print(
    "\nCODE CELL 04 STATUS:",
    "STRATUM_SIGNAL_INSPECTION_COMPLETE"
)
```

<h4>معیار پایان این Inspection</h4>

<ul>
<li>Schema واقعی Candidateهای Chunking شناخته شود.</li>
<li>Lineage بین Chunk و Source Document قابل مشاهده باشد.</li>
<li>وجود یا نبود Signal مستقیم برای Split/Fallback/Oversize مشخص شود.</li>
<li>Token Countهای ذخیره‌شده و Speaker Composition در صورت وجود شناسایی شوند.</li>
<li>مشخص شود Long-Utterance Strata را می‌توان مستقیماً از Artifactها ساخت یا نیاز به Measurement مستقل دارد.</li>
</ul>

```python
# Code Cell 05

def flatten_non_text_fields(
    obj,
    prefix="",
    output=None,
    max_value_chars=300,
):
    if output is None:
        output = []

    if isinstance(obj, dict):
        for key, value in obj.items():
            path = f"{prefix}.{key}" if prefix else str(key)

            # Avoid printing chunk/source text payloads.
            if key.lower() in {
                "text",
                "content",
                "source_text",
                "chunk_text",
                "parent_text",
                "child_text",
            }:
                continue

            if isinstance(value, (dict, list)):
                flatten_non_text_fields(
                    value,
                    prefix=path,
                    output=output,
                    max_value_chars=max_value_chars,
                )
            else:
                rendered = scalarize(value)

                if len(rendered) > max_value_chars:
                    rendered = (
                        rendered[:max_value_chars]
                        + "...<TRUNCATED>"
                    )

                output.append((path, rendered))

    elif isinstance(obj, list):
        for item in obj:
            path = f"{prefix}[]"

            if isinstance(item, (dict, list)):
                flatten_non_text_fields(
                    item,
                    prefix=path,
                    output=output,
                    max_value_chars=max_value_chars,
                )
            else:
                rendered = scalarize(item)

                if len(rendered) > max_value_chars:
                    rendered = (
                        rendered[:max_value_chars]
                        + "...<TRUNCATED>"
                    )

                output.append((path, rendered))

    return output


signal_terms = {
    "token",
    "split",
    "fallback",
    "sentence",
    "oversize",
    "utterance",
    "document",
    "source_id",
    "source_document",
    "lineage",
    "speaker",
    "composition",
    "locator",
    "parent",
    "child",
    "boundary",
}


print("=== EXACT TRANSCRIPT CHUNK CONTRACT ===")

chunk_contract = parsed_json["chunk_contract"]

builder_specification = (
    chunk_contract.get("builder_specification") or {}
)

transcript_contract = (
    builder_specification.get("transcript")
)

print(
    json.dumps(
        transcript_contract,
        ensure_ascii=False,
        indent=2,
    )
)


print("\n=== FROZEN CHUNK ARTIFACT INSPECTION ===")

strategy_schemas = {}
strategy_nested_paths = defaultdict(Counter)
strategy_signal_values = defaultdict(
    lambda: defaultdict(Counter)
)
strategy_source_types = defaultdict(Counter)
strategy_rows = Counter()
strategy_examples = {}


for strategy_name, path in candidate_paths.items():

    print(f"\n--- {strategy_name} ---")
    print("path:", path.relative_to(repo_root))
    print("exists:", path.is_file())

    if not path.is_file():
        continue

    top_level_keys = set()

    with path.open("r", encoding="utf-8") as f:
        for line_number, line in enumerate(f, start=1):

            row = json.loads(line)
            strategy_rows[strategy_name] += 1

            top_level_keys.update(row.keys())

            source_type = row.get("source_type")

            if source_type is not None:
                strategy_source_types[
                    strategy_name
                ][str(source_type)] += 1

            flat_fields = flatten_non_text_fields(row)

            for field_path, value in flat_fields:
                strategy_nested_paths[
                    strategy_name
                ][field_path] += 1

                lowered = field_path.lower()

                if any(
                    term in lowered
                    for term in signal_terms
                ):
                    strategy_signal_values[
                        strategy_name
                    ][field_path][value] += 1

            if (
                source_type == "transcript"
                and strategy_name not in strategy_examples
            ):
                strategy_examples[strategy_name] = {
                    field_path: value
                    for field_path, value in flat_fields
                }

    strategy_schemas[strategy_name] = sorted(
        top_level_keys
    )

    print(
        "rows:",
        strategy_rows[strategy_name]
    )

    print(
        "top_level_keys:",
        strategy_schemas[strategy_name]
    )

    print(
        "source_type_distribution:",
        dict(
            strategy_source_types[strategy_name]
        )
    )


print("\n\n=== SIGNAL-BEARING FIELD PATHS ===")

for strategy_name in candidate_paths:

    print(f"\n[{strategy_name}]")

    path_map = strategy_signal_values[
        strategy_name
    ]

    if not path_map:
        print("NO_SIGNAL_PATHS")
        continue

    for field_path in sorted(path_map):
        counts = path_map[field_path]

        print(
            field_path,
            "->",
            counts.most_common(12)
        )


print("\n\n=== REPRESENTATIVE TRANSCRIPT CHUNK METADATA ===")

for strategy_name in candidate_paths:

    print(f"\n[{strategy_name}]")

    example = strategy_examples.get(
        strategy_name
    )

    if example is None:
        print(
            "NO_TRANSCRIPT_EXAMPLE_FOUND"
        )
        continue

    for field_path in sorted(example):
        lowered = field_path.lower()

        if any(
            term in lowered
            for term in signal_terms
        ):
            print(
                field_path,
                "=",
                example[field_path]
            )


print("\n\n=== CROSS-STRATEGY FIELD DISCOVERY ===")

all_field_paths = set()

for strategy_name, path_counts in (
    strategy_nested_paths.items()
):
    all_field_paths.update(
        path_counts.keys()
    )

candidate_mapping_fields = sorted(
    field_path
    for field_path in all_field_paths
    if any(
        term in field_path.lower()
        for term in {
            "document_id",
            "source_document",
            "source_ids",
            "source_id",
            "utterance",
            "lineage",
            "locator",
            "token",
            "split",
            "fallback",
            "sentence",
            "oversize",
        }
    )
)

for field_path in candidate_mapping_fields:
    present_in = [
        strategy_name
        for strategy_name
        in candidate_paths
        if field_path
        in strategy_nested_paths[
            strategy_name
        ]
    ]

    print(
        field_path,
        "->",
        present_in
    )


print("\n=== INSPECTION FLAGS ===")

all_strategy_files_read = all(
    strategy_rows[name] > 0
    for name in candidate_paths
)

has_document_mapping_signal = any(
    any(
        term in field_path.lower()
        for term in {
            "document_id",
            "source_document",
            "source_ids",
            "utterance",
            "lineage",
        }
    )
    for field_path in all_field_paths
)

has_token_signal = any(
    "token" in field_path.lower()
    for field_path in all_field_paths
)

has_split_signal = any(
    any(
        term in field_path.lower()
        for term in {
            "split",
            "fallback",
            "sentence",
            "oversize",
        }
    )
    for field_path in all_field_paths
)

inspection_flags = {
    "all_frozen_candidate_files_read":
        all_strategy_files_read,

    "source_mapping_signal_present":
        has_document_mapping_signal,

    "stored_token_signal_present":
        has_token_signal,

    "stored_split_or_fallback_signal_present":
        has_split_signal,
}

for key, value in inspection_flags.items():
    print(
        key,
        ":",
        "YES" if value else "NO"
    )


print(
    "\nCODE CELL 05 STATUS:",
    "FROZEN_CHUNK_SCHEMA_INSPECTION_COMPLETE"
)
```

<li>اگر جزئیات لازم در Notebook موجود نباشد، این محدودیت صریحاً ثبت شود و چیزی حدس زده نشود.</li>

```python
# Code Cell 06

nb11_path = required_paths["nb11_notebook"]

with nb11_path.open("r", encoding="utf-8") as f:
    nb11 = json.load(f)

code_cells = [
    cell
    for cell in nb11.get("cells", [])
    if cell.get("cell_type") == "code"
]

search_groups = {
    "oversize_utterance": [
        "oversize_utterance",
        "transcript_oversize_utterance_fragment",
    ],
    "sentence_logic": [
        "sentence_boundary",
        "sentence",
        "split_sent",
    ],
    "token_offset_fallback": [
        "token_offset",
        "TOKEN_OFFSET_WINDOW",
        "fallback",
    ],
    "token_measurement": [
        "multilingual-e5-small",
        "token_count",
        "tokenizer",
    ],
}


def cell_source(cell):
    source = cell.get("source", [])

    if isinstance(source, list):
        return "".join(source)

    return str(source)


def context_around_matches(
    source,
    terms,
    context_lines=8,
):
    lines = source.splitlines()
    hit_indexes = []

    lowered_terms = [
        term.lower()
        for term in terms
    ]

    for index, line in enumerate(lines):
        lowered_line = line.lower()

        if any(
            term in lowered_line
            for term in lowered_terms
        ):
            hit_indexes.append(index)

    if not hit_indexes:
        return []

    ranges = []

    for index in hit_indexes:
        start = max(
            0,
            index - context_lines
        )

        end = min(
            len(lines),
            index + context_lines + 1
        )

        ranges.append(
            (start, end)
        )

    merged = []

    for start, end in ranges:
        if (
            not merged
            or start > merged[-1][1]
        ):
            merged.append(
                [start, end]
            )
        else:
            merged[-1][1] = max(
                merged[-1][1],
                end
            )

    excerpts = []

    for start, end in merged:
        excerpt = "\n".join(
            f"{line_no + 1:04d}: {lines[line_no]}"
            for line_no in range(start, end)
        )

        excerpts.append(excerpt)

    return excerpts


group_hits = defaultdict(list)

for cell_index, cell in enumerate(
    code_cells,
    start=1,
):
    source = cell_source(cell)

    for group_name, terms in search_groups.items():
        excerpts = context_around_matches(
            source,
            terms,
        )

        if excerpts:
            group_hits[group_name].append(
                {
                    "code_cell_position": cell_index,
                    "excerpts": excerpts,
                }
            )


print("=== NB11 IMPLEMENTATION DISCOVERY ===")
print("NB11 path:", nb11_path.relative_to(repo_root))
print("Code cells:", len(code_cells))


for group_name in search_groups:

    print(f"\n\n=== {group_name.upper()} ===")

    hits = group_hits.get(
        group_name,
        []
    )

    print(
        "matching code cells:",
        len(hits)
    )

    if not hits:
        print(
            "NO_IMPLEMENTATION_MATCH_FOUND"
        )
        continue

    for hit in hits:
        print(
            "\n--- NB11 code cell position",
            hit["code_cell_position"],
            "---"
        )

        for excerpt_number, excerpt in enumerate(
            hit["excerpts"],
            start=1,
        ):
            print(
                f"\n[excerpt {excerpt_number}]"
            )
            print(excerpt)


print("\n\n=== FUNCTION / CONSTANT DISCOVERY ===")

interesting_definition_lines = []

definition_terms = {
    "oversize",
    "sentence",
    "token",
    "transcript",
    "atomic",
    "split",
}

for cell_index, cell in enumerate(
    code_cells,
    start=1,
):
    source = cell_source(cell)

    for line in source.splitlines():
        stripped = line.strip()
        lowered = stripped.lower()

        is_definition = (
            stripped.startswith("def ")
            or stripped.startswith("class ")
            or (
                "=" in stripped
                and not stripped.startswith("#")
            )
        )

        if (
            is_definition
            and any(
                term in lowered
                for term in definition_terms
            )
        ):
            interesting_definition_lines.append(
                (
                    cell_index,
                    stripped,
                )
            )


for cell_index, line in interesting_definition_lines:
    print(
        f"NB11 code cell {cell_index}:",
        line
    )


implementation_flags = {
    "oversize_fragment_marker_found":
        bool(
            group_hits["oversize_utterance"]
        ),

    "sentence_logic_found":
        bool(
            group_hits["sentence_logic"]
        ),

    "token_offset_fallback_logic_found":
        bool(
            group_hits["token_offset_fallback"]
        ),

    "token_measurement_logic_found":
        bool(
            group_hits["token_measurement"]
        ),
}


print("\n=== IMPLEMENTATION FLAGS ===")

for name, value in implementation_flags.items():
    print(
        name,
        ":",
        "YES" if value else "NO"
    )


implementation_complete = all(
    implementation_flags.values()
)

print(
    "\nCODE CELL 06 GATE:",
    "PASS"
    if implementation_complete
    else "FAIL"
)
```

<h3>Operational Stratum Rule Registry</h3>

<p>
Golden Dataset Contract با موفقیت Validate شد. اکنون هر یک از Strataهای اجباری NB12
به یک Rule عملیاتی، قابل Audit و قابل‌بازتولید تبدیل می‌شود.
</p>

<p>
اصل این مرحله این است که <strong>نام یک Stratum به‌تنهایی کافی نیست</strong>.
برای ورود یک Case به Stratum باید مشخص باشد چه Evidence ساختاری، Metadata، Token Measurement
یا Review انسانی آن عضویت را اثبات می‌کند.
</p>

<p>
برای مفاهیمی مانند «کوتاه» و «بلند» که Threshold رسمی از قبل Freeze نشده است،
به‌جای عدد دلخواه از Distribution واقعی همان Population استفاده می‌شود:
ربع پایین برای Short و ربع بالا برای Long.
این Thresholdها در همین مرحله از Corpus واقعی محاسبه می‌شوند و بعداً همراه Contract ثبت خواهند شد.
</p>

<p>
Strataهایی که ذاتاً Semantic هستند، مانند <code>multi_topic_article</code>،
<code>multi_paragraph_evidence_case</code> و <code>question_answer_transition</code>،
صرفاً با یک Signal ضعیف به‌طور خودکار تأیید نمی‌شوند.
برای آن‌ها Structural Prefilter مجاز است، اما عضویت نهایی نیازمند Evidence متنی یا Human Review خواهد بود.
</p>

<p>
این Cell فقط Rule Registry را در حافظه تعریف و Validate می‌کند.
هنوز هیچ Golden Case انتخاب، Sample یا تولید نمی‌شود.
</p>

```python
# Code Cell 08

# ============================================================
# Evidence-derived thresholds
# ============================================================

website_direct_lengths = (
    website_origin_lengths[
        "primary_author"
    ]
)

website_supervised_lengths = (
    website_origin_lengths[
        "supervised"
    ]
)

assert website_direct_lengths, (
    "No direct-author Website population available."
)

assert website_supervised_lengths, (
    "No supervised Website population available."
)


WEBSITE_LENGTH_BANDS_V1 = {
    "direct_author": {
        "population":
            len(
                website_direct_lengths
            ),

        "short_max_chars":
            percentile(
                website_direct_lengths,
                0.25,
            ),

        "long_min_chars":
            percentile(
                website_direct_lengths,
                0.75,
            ),

        "measurement":
            "SOURCE_TEXT_CHAR_COUNT",

        "threshold_policy":
            "WITHIN_AUTHORITY_GROUP_EMPIRICAL_QUARTILES",
    },

    "supervised": {
        "population":
            len(
                website_supervised_lengths
            ),

        "short_max_chars":
            percentile(
                website_supervised_lengths,
                0.25,
            ),

        "long_min_chars":
            percentile(
                website_supervised_lengths,
                0.75,
            ),

        "measurement":
            "SOURCE_TEXT_CHAR_COUNT",

        "threshold_policy":
            "WITHIN_AUTHORITY_GROUP_EMPIRICAL_QUARTILES",
    },
}


positive_book_heading_depths = []

for depth, count in (
    book_heading_depth.items()
):
    if depth <= 0:
        continue

    positive_book_heading_depths.extend(
        [depth] * count
    )

assert positive_book_heading_depths, (
    "No positive Book heading depth found."
)

deep_heading_threshold = int(
    percentile(
        positive_book_heading_depths,
        0.90,
    )
)

deep_heading_threshold = max(
    1,
    deep_heading_threshold,
)


# ============================================================
# Operational Rule Registry
# ============================================================

STRATUM_RULES_V1_DRAFT = {

    # --------------------------------------------------------
    # WEBSITE
    # --------------------------------------------------------

    "website": {

        "direct_author_short": {
            "eligibility":
                "AUTO_ELIGIBLE",

            "unit":
                "SOURCE_DOCUMENT",

            "conditions": {
                "source_type":
                    "website",

                "content_origin":
                    "primary_author",

                "content_authority_type":
                    "direct_author",

                "text_char_count_lte":
                    WEBSITE_LENGTH_BANDS_V1[
                        "direct_author"
                    ][
                        "short_max_chars"
                    ],
            },

            "evidence_basis":
                "ACCEPTED_CORPUS_METADATA_AND_EMPIRICAL_LENGTH_BAND",

            "review_requirement":
                "NONE_FOR_STRATUM_MEMBERSHIP",
        },


        "direct_author_long": {
            "eligibility":
                "AUTO_ELIGIBLE",

            "unit":
                "SOURCE_DOCUMENT",

            "conditions": {
                "source_type":
                    "website",

                "content_origin":
                    "primary_author",

                "content_authority_type":
                    "direct_author",

                "text_char_count_gte":
                    WEBSITE_LENGTH_BANDS_V1[
                        "direct_author"
                    ][
                        "long_min_chars"
                    ],
            },

            "evidence_basis":
                "ACCEPTED_CORPUS_METADATA_AND_EMPIRICAL_LENGTH_BAND",

            "review_requirement":
                "NONE_FOR_STRATUM_MEMBERSHIP",
        },


        "supervised_short": {
            "eligibility":
                "AUTO_ELIGIBLE",

            "unit":
                "SOURCE_DOCUMENT",

            "conditions": {
                "source_type":
                    "website",

                "content_origin":
                    "supervised",

                "content_authority_type":
                    "editorial_supervision",

                "text_char_count_lte":
                    WEBSITE_LENGTH_BANDS_V1[
                        "supervised"
                    ][
                        "short_max_chars"
                    ],
            },

            "evidence_basis":
                "ACCEPTED_CORPUS_METADATA_AND_EMPIRICAL_LENGTH_BAND",

            "review_requirement":
                "NONE_FOR_STRATUM_MEMBERSHIP",
        },


        "supervised_long": {
            "eligibility":
                "AUTO_ELIGIBLE",

            "unit":
                "SOURCE_DOCUMENT",

            "conditions": {
                "source_type":
                    "website",

                "content_origin":
                    "supervised",

                "content_authority_type":
                    "editorial_supervision",

                "text_char_count_gte":
                    WEBSITE_LENGTH_BANDS_V1[
                        "supervised"
                    ][
                        "long_min_chars"
                    ],
            },

            "evidence_basis":
                "ACCEPTED_CORPUS_METADATA_AND_EMPIRICAL_LENGTH_BAND",

            "review_requirement":
                "NONE_FOR_STRATUM_MEMBERSHIP",
        },


        "multi_topic_article": {
            "eligibility":
                "HUMAN_REVIEW_REQUIRED",

            "unit":
                "SOURCE_DOCUMENT",

            "candidate_prefilter": {
                "source_type":
                    "website",

                "signal":
                    "MULTIPLE_DISTINCT_SOURCE_TAGS_OR_MULTIPLE_TOPIC_EVIDENCE",
            },

            "acceptance_rule": (
                "AT_LEAST_TWO_SEMANTICALLY_DISTINCT_FINANCIAL_TOPICS_"
                "SUPPORTED_BY_ARTICLE_TEXT"
            ),

            "forbidden_shortcut":
                "MULTIPLE_TAGS_ALONE_DO_NOT_PROVE_MULTI_TOPIC",

            "evidence_basis":
                "SOURCE_METADATA_PREFILTER_PLUS_TEXTUAL_REVIEW",
        },


        "citation_locator_sensitive_case": {
            "eligibility":
                "HUMAN_REVIEW_REQUIRED",

            "unit":
                "SOURCE_DOCUMENT",

            "candidate_prefilter": {
                "source_type":
                    "website",

                "required_locator_fields": [
                    "document_id",
                    "canonical_url_or_url",
                ],

                "preferred_signal": (
                    "SEMANTICALLY_SIMILAR_OR_DUPLICATE_FAMILY_WITH_"
                    "DISTINCT_SOURCE_LOCATORS"
                ),
            },

            "acceptance_rule": (
                "CORRECT_ANSWER_REQUIRES_RECOVERING_THE_EXACT_"
                "SOURCE_DOCUMENT_LOCATOR_NOT_ONLY_A_SIMILAR_ARTICLE"
            ),

            "evidence_basis":
                "NB09_DUPLICATE_OR_SIMILARITY_EVIDENCE_PLUS_EXACT_LOCATOR",
        },
    },


    # --------------------------------------------------------
    # BOOK
    # --------------------------------------------------------

    "book": {

        "intrinsically_tiny_narrative_region": {
            "eligibility":
                "AUTO_ELIGIBLE",

            "unit":
                "PROTECTED_STRUCTURAL_REGION",

            "conditions": {
                "source_type":
                    "book",

                "structural_type":
                    "book_paragraph_group",

                "whole_protected_region_tokens_lt":
                    100,
            },

            "token_basis":
                "FROZEN_NB11_MEASUREMENT_TOKENIZER",

            "evidence_basis":
                "NB11_ROOT_CAUSE_REGION_EVIDENCE",

            "interpretation":
                "STRUCTURAL_TINY_REGION_NOT_BUILDER_FAILURE",
        },


        "intrinsically_tiny_table_region": {
            "eligibility":
                "AUTO_ELIGIBLE",

            "unit":
                "PROTECTED_TABLE_REGION",

            "conditions": {
                "source_type":
                    "book",

                "structural_type":
                    "table_record",

                "whole_protected_region_tokens_lt":
                    100,
            },

            "token_basis":
                "FROZEN_NB11_MEASUREMENT_TOKENIZER",

            "evidence_basis":
                "NB11_ROOT_CAUSE_REGION_EVIDENCE",
        },


        "long_section": {
            "eligibility":
                "AUTO_ELIGIBLE",

            "unit":
                "BOOK_SECTION",

            "conditions": {
                "source_type":
                    "book",

                "section_token_count_gt":
                    800,
            },

            "threshold_reason": (
                "EXCEEDS_STRUCTURE_AWARE_MEDIUM_HARD_MAXIMUM_"
                "AND_TESTS_MULTI_CHUNK_EVIDENCE"
            ),

            "token_basis":
                "FROZEN_NB11_MEASUREMENT_TOKENIZER",

            "evidence_basis":
                "NB11_BOOK_SECTION_CALIBRATION",
        },


        "large_table_run": {
            "eligibility":
                "AUTO_ELIGIBLE",

            "unit":
                "CONTIGUOUS_TABLE_RUN",

            "conditions": {
                "source_type":
                    "book",

                "table_run_token_count_gt":
                    500,
            },

            "threshold_reason": (
                "EXCEEDS_SMALL_CHILD_HARD_MAXIMUM_AND_"
                "STRESSES_TABLE_PRESERVATION"
            ),

            "token_basis":
                "FROZEN_NB11_MEASUREMENT_TOKENIZER",

            "evidence_basis":
                "NB11_TABLE_RUN_CALIBRATION",
        },


        "exercise_or_self_assessment": {
            "eligibility":
                "AUTO_ELIGIBLE_WITH_EXPLICIT_MARKER",

            "unit":
                "BOOK_REGION",

            "conditions": {
                "source_type":
                    "book",

                "explicit_heading_or_style_marker_any": [
                    "خودارزیابی",
                    "کاربرگ",
                    "تمرین",
                ],
            },

            "acceptance_rule":
                "EXPLICIT_EDUCATIONAL_ACTIVITY_MARKER_MUST_BE_PRESENT",

            "semantic_inference_without_marker":
                "FORBIDDEN",

            "evidence_basis":
                "BOOK_HEADING_PATH_OR_EXPLICIT_STYLE_TEXT",
        },


        "supplemental_footnote_case": {
            "eligibility":
                "AUTO_ELIGIBLE",

            "unit":
                "PARENT_PLUS_SUPPLEMENTAL_RECORD",

            "conditions": {
                "source_type":
                    "book",

                "content_id_present":
                    True,

                "parent_paragraph_id_present":
                    True,
            },

            "evidence_basis":
                "BOOK_SUPPLEMENTAL_LOCATOR_LINKAGE",

            "parent_child_link_must_be_recoverable":
                True,
        },


        "deep_heading_path": {
            "eligibility":
                "AUTO_ELIGIBLE",

            "unit":
                "BOOK_RECORD",

            "conditions": {
                "source_type":
                    "book",

                "heading_depth_gte":
                    deep_heading_threshold,
            },

            "threshold_policy":
                "OBSERVED_BOOK_HEADING_DEPTH_P90",

            "evidence_basis":
                "BOOK_LOCATOR_HEADING_PATH",
        },


        "multi_paragraph_evidence_case": {
            "eligibility":
                "HUMAN_REVIEW_REQUIRED",

            "unit":
                "BOOK_SECTION",

            "candidate_prefilter": {
                "source_type":
                    "book",

                "same_heading_region":
                    True,

                "minimum_distinct_paragraphs":
                    2,
            },

            "acceptance_rule": (
                "ANSWER_REQUIRES_COMPLEMENTARY_EVIDENCE_FROM_"
                "AT_LEAST_TWO_DISTINCT_SOURCE_PARAGRAPHS"
            ),

            "single_paragraph_answerable_case":
                "REJECT_FROM_THIS_STRATUM",

            "evidence_basis":
                "STRUCTURAL_PREFILTER_PLUS_TEXTUAL_REVIEW",
        },
    },


    # --------------------------------------------------------
    # TRANSCRIPT
    # --------------------------------------------------------

    "transcript": {

        "single_speaker_primary": {
            "eligibility":
                "AUTO_ELIGIBLE",

            "unit":
                "PHYSICAL_FILE",

            "conditions": {
                "source_type":
                    "transcript",

                "effective_speaker_count":
                    1,

                "all_primary_dr_roudi_corpus":
                    True,
            },

            "evidence_basis":
                "TRANSCRIPT_EFFECTIVE_SPEAKER_AND_PRIMARY_SIGNAL",
        },


        "single_speaker_unresolved": {
            "eligibility":
                "AUTO_ELIGIBLE",

            "unit":
                "PHYSICAL_FILE",

            "conditions": {
                "source_type":
                    "transcript",

                "effective_speaker_count":
                    1,

                "all_base_status":
                    "UNRESOLVED",
            },

            "unknown_speaker_inference":
                "FORBIDDEN",

            "evidence_basis":
                "TRANSCRIPT_BASE_SPEAKER_STATUS",
        },


        "multi_speaker_uni_tehran": {
            "eligibility":
                "AUTO_ELIGIBLE",

            "unit":
                "PHYSICAL_FILE",

            "conditions": {
                "program_id":
                    "program:uni-tehran",

                "effective_speaker_count_gt":
                    1,
            },

            "evidence_basis":
                "TRANSCRIPT_EFFECTIVE_SPEAKER_COMPOSITION",
        },


        "multi_speaker_chehelston": {
            "eligibility":
                "AUTO_ELIGIBLE",

            "unit":
                "PHYSICAL_FILE",

            "conditions": {
                "program_id":
                    "program:chehelston",

                "effective_speaker_count_gt":
                    1,
            },

            "evidence_basis":
                "TRANSCRIPT_EFFECTIVE_SPEAKER_COMPOSITION",
        },


        "multi_speaker_asre_shirin": {
            "eligibility":
                "AUTO_ELIGIBLE",

            "unit":
                "PHYSICAL_FILE",

            "conditions": {
                "program_id":
                    "program:asre-shirin",

                "effective_speaker_count_gt":
                    1,
            },

            "evidence_basis":
                "TRANSCRIPT_EFFECTIVE_SPEAKER_COMPOSITION",
        },


        "multi_speaker_khane_to": {
            "eligibility":
                "AUTO_ELIGIBLE",

            "unit":
                "PHYSICAL_FILE",

            "conditions": {
                "program_id":
                    "program:khane-to",

                "effective_speaker_count_gt":
                    1,
            },

            "evidence_basis":
                "TRANSCRIPT_EFFECTIVE_SPEAKER_COMPOSITION",
        },


        "multi_speaker_mizan": {
            "eligibility":
                "AUTO_ELIGIBLE",

            "unit":
                "PHYSICAL_FILE",

            "conditions": {
                "program_id":
                    "program:mizan",

                "effective_speaker_count_gt":
                    1,
            },

            "evidence_basis":
                "TRANSCRIPT_EFFECTIVE_SPEAKER_COMPOSITION",
        },


        "primary_to_nonprimary_transition": {
            "eligibility":
                "AUTO_ELIGIBLE",

            "unit":
                "ADJACENT_UTTERANCE_PAIR",

            "conditions": {
                "same_physical_file":
                    True,

                "left_primary_dr_roudi_corpus":
                    True,

                "right_primary_dr_roudi_corpus":
                    False,
            },

            "important_exclusion": (
                "UNRESOLVED_OR_NULL_PRIMARY_SIGNAL_IS_NOT_"
                "RELABELED_AS_CONFIRMED_NONPRIMARY"
            ),

            "evidence_basis":
                "EXPLICIT_ADJACENT_RECORD_PRIMARY_SIGNAL",
        },


        "mixed_primary_signal_case": {
            "eligibility":
                "AUTO_ELIGIBLE",

            "unit":
                "PHYSICAL_FILE",

            "conditions": {
                "distinct_primary_signal_states_gt":
                    1,
            },

            "allowed_observed_states": [
                True,
                False,
                None,
            ],

            "evidence_basis":
                "TRANSCRIPT_FILE_LEVEL_PRIMARY_SIGNAL_COMPOSITION",
        },


        "long_utterance_over_500": {
            "eligibility":
                "AUTO_ELIGIBLE",

            "unit":
                "SOURCE_UTTERANCE",

            "conditions": {
                "source_type":
                    "transcript",

                "source_utterance_tokens_gt":
                    500,
            },

            "token_basis":
                "FROZEN_NB11_MEASUREMENT_TOKENIZER",

            "evidence_basis":
                "NB11_SOURCE_UTTERANCE_TOKEN_MEASUREMENT",
        },


        "long_utterance_over_800": {
            "eligibility":
                "AUTO_ELIGIBLE",

            "unit":
                "SOURCE_UTTERANCE",

            "conditions": {
                "source_type":
                    "transcript",

                "source_utterance_tokens_gt":
                    800,
            },

            "token_basis":
                "FROZEN_NB11_MEASUREMENT_TOKENIZER",

            "evidence_basis":
                "NB11_SOURCE_UTTERANCE_TOKEN_MEASUREMENT",
        },


        "token_fallback_sentence_case": {
            "eligibility":
                "AUTO_ELIGIBLE_WITH_EXACT_FALLBACK_EVIDENCE",

            "unit":
                "SOURCE_SENTENCE",

            "conditions": {
                "source_type":
                    "transcript",

                "sentence_tokens_gt":
                    500,

                "small_or_child_fragment_marker":
                    "transcript_token_fallback_fragment",
            },

            "split_order": [
                "SENTENCE_BOUNDARY",
                "TOKEN_OFFSET_WINDOW_FOR_OVERSIZE_SENTENCE_ONLY",
            ],

            "generic_oversize_fragment_alone_is_sufficient":
                False,

            "token_basis":
                "FROZEN_NB11_MEASUREMENT_TOKENIZER",

            "evidence_basis":
                "NB11_EXACT_FALLBACK_IMPLEMENTATION",
        },


        "question_answer_transition": {
            "eligibility":
                "HUMAN_REVIEW_REQUIRED",

            "unit":
                "ADJACENT_OR_LOCAL_TRANSCRIPT_CONTEXT",

            "candidate_prefilter": {
                "same_physical_file":
                    True,

                "speaker_transition_preferred":
                    True,
            },

            "acceptance_rule": (
                "TEXT_MUST_EXPLICITLY_SUPPORT_A_QUESTION_TO_ANSWER_"
                "RELATIONSHIP_OR_HUMAN_REVIEW_MUST_CONFIRM_IT"
            ),

            "speaker_transition_alone":
                "INSUFFICIENT",

            "direct_qa_metadata_available":
                False,

            "unknown_speaker_inference":
                "FORBIDDEN",

            "evidence_basis":
                "SOURCE_TEXT_PLUS_HUMAN_REVIEW",
        },
    },
}


# ============================================================
# Registry Validation
# ============================================================

expected_strata = {
    source_type:
        list(
            handoff[source_type]
        )
    for source_type
    in [
        "website",
        "book",
        "transcript",
    ]
}

actual_strata = {
    source_type:
        list(
            STRATUM_RULES_V1_DRAFT[
                source_type
            ].keys()
        )
    for source_type
    in [
        "website",
        "book",
        "transcript",
    ]
}


strata_exact_match_gate = (
    actual_strata
    == expected_strata
)


total_expected_strata = sum(
    len(values)
    for values
    in expected_strata.values()
)

total_actual_strata = sum(
    len(values)
    for values
    in actual_strata.values()
)

strata_count_gate = (
    total_expected_strata
    == 27

    and

    total_actual_strata
    == 27
)


all_rules_have_eligibility_gate = all(
    isinstance(rule, dict)
    and bool(
        rule.get(
            "eligibility"
        )
    )

    for source_rules
    in STRATUM_RULES_V1_DRAFT.values()

    for rule
    in source_rules.values()
)


unknown_inference_guardrail_gate = (
    STRATUM_RULES_V1_DRAFT[
        "transcript"
    ][
        "single_speaker_unresolved"
    ][
        "unknown_speaker_inference"
    ]
    == "FORBIDDEN"

    and

    STRATUM_RULES_V1_DRAFT[
        "transcript"
    ][
        "question_answer_transition"
    ][
        "unknown_speaker_inference"
    ]
    == "FORBIDDEN"
)


qa_review_gate = (
    STRATUM_RULES_V1_DRAFT[
        "transcript"
    ][
        "question_answer_transition"
    ][
        "eligibility"
    ]
    == "HUMAN_REVIEW_REQUIRED"

    and

    STRATUM_RULES_V1_DRAFT[
        "transcript"
    ][
        "question_answer_transition"
    ][
        "speaker_transition_alone"
    ]
    == "INSUFFICIENT"
)


semantic_review_gate = all(
    STRATUM_RULES_V1_DRAFT[
        source_type
    ][
        stratum
    ][
        "eligibility"
    ]
    == "HUMAN_REVIEW_REQUIRED"

    for source_type, stratum
    in [
        (
            "website",
            "multi_topic_article",
        ),
        (
            "website",
            "citation_locator_sensitive_case",
        ),
        (
            "book",
            "multi_paragraph_evidence_case",
        ),
        (
            "transcript",
            "question_answer_transition",
        ),
    ]
)


length_threshold_gate = (
    WEBSITE_LENGTH_BANDS_V1[
        "direct_author"
    ][
        "short_max_chars"
    ]
    <=
    WEBSITE_LENGTH_BANDS_V1[
        "direct_author"
    ][
        "long_min_chars"
    ]

    and

    WEBSITE_LENGTH_BANDS_V1[
        "supervised"
    ][
        "short_max_chars"
    ]
    <=
    WEBSITE_LENGTH_BANDS_V1[
        "supervised"
    ][
        "long_min_chars"
    ]
)


no_sampling_gate = (
    GOLDEN_DATASET_CONTRACT_V1_DRAFT[
        "sampling_policy"
    ][
        "status"
    ]
    == "NOT_YET_DEFINED"
)


no_question_generation_gate = (
    GOLDEN_DATASET_CONTRACT_V1_DRAFT[
        "question_answer_policy"
    ][
        "question_generation_status"
    ]
    == "NOT_STARTED"
)


rule_gates = {
    "exact_handoff_strata_match":
        strata_exact_match_gate,

    "exact_27_strata":
        strata_count_gate,

    "every_rule_has_eligibility":
        all_rules_have_eligibility_gate,

    "unknown_speaker_guardrail":
        unknown_inference_guardrail_gate,

    "qa_requires_review":
        qa_review_gate,

    "semantic_strata_require_review":
        semantic_review_gate,

    "website_length_thresholds_valid":
        length_threshold_gate,

    "sampling_not_started":
        no_sampling_gate,

    "question_generation_not_started":
        no_question_generation_gate,
}


rules_pass = all(
    rule_gates.values()
)


print("=== EMPIRICAL WEBSITE LENGTH BANDS ===")

print(
    json.dumps(
        WEBSITE_LENGTH_BANDS_V1,
        ensure_ascii=False,
        indent=2,
    )
)


print("\n=== BOOK DEEP-HEADING THRESHOLD ===")

print(
    "Observed positive heading records:",
    len(
        positive_book_heading_depths
    )
)

print(
    "Observed maximum depth:",
    max(
        positive_book_heading_depths
    )
)

print(
    "P90 operational deep-heading threshold:",
    deep_heading_threshold
)


print("\n=== STRATUM RULE SUMMARY ===")

for source_type in [
    "website",
    "book",
    "transcript",
]:

    print(
        f"\n[{source_type.upper()}]"
    )

    for stratum, rule in (
        STRATUM_RULES_V1_DRAFT[
            source_type
        ].items()
    ):
        print(
            "-",
            stratum,
            "| eligibility:",
            rule[
                "eligibility"
            ],
            "| unit:",
            rule[
                "unit"
            ],
        )


print("\n=== RULE REGISTRY GATES ===")

for name, passed in (
    rule_gates.items()
):
    print(
        name,
        ":",
        "PASS" if passed else "FAIL"
    )


print(
    "\nCODE CELL 08 GATE:",
    "PASS" if rules_pass else "FAIL"
)


assert rules_pass, (
    "Operational stratum rule registry validation failed. "
    "Do not discover or sample Golden candidates until "
    "all required strata have explicit defensible rules."
)
```

```python
# Code Cell 09

# ============================================================
# Candidate Discovery — Stage A
# Repair:
# Zero-match exact rules are evidence, not automatic failure.
# They must be surfaced for targeted rule review.
# ============================================================


def stable_json_key(value):
    return json.dumps(
        value,
        ensure_ascii=False,
        sort_keys=True,
        default=str,
    )


def heading_depth_from_value(value):
    if isinstance(value, list):
        return len(value)

    if isinstance(value, str) and value.strip():
        return len(
            [
                part
                for part in re.split(
                    r"\s*(?:>|/|→|\||»)\s*",
                    value,
                )
                if part.strip()
            ]
        )

    return 0


def explicit_activity_marker(row):
    metadata = row.get("source_metadata") or {}
    locator = row.get("locator") or {}

    heading_path = (
        locator.get("heading_path")
        or metadata.get("heading_path")
    )

    style = metadata.get("style")
    text = str(row.get("text") or "")

    marker_terms = [
        "خودارزیابی",
        "کاربرگ",
        "تمرین",
    ]

    searchable_parts = [
        stable_json_key(heading_path),
        str(style or ""),
    ]

    if metadata.get("is_heading") is True:
        searchable_parts.append(text)

    searchable = "\n".join(
        searchable_parts
    )

    return [
        term
        for term in marker_terms
        if term in searchable
    ]


def normalized_tags(row):
    metadata = row.get("source_metadata") or {}
    tags = metadata.get("tags")

    if tags is None:
        return []

    if isinstance(tags, list):
        values = tags
    elif isinstance(tags, (tuple, set)):
        values = list(tags)
    else:
        values = [tags]

    normalized = []

    for value in values:

        if isinstance(value, dict):
            rendered = str(
                value.get("name")
                or ""
            ).strip()
        else:
            rendered = str(
                value
                or ""
            ).strip()

        if rendered:
            normalized.append(
                rendered
            )

    return sorted(
        set(normalized)
    )


def compact_example(
    source_type,
    stratum,
    item,
):
    if source_type == "website":
        return {
            key: item.get(key)
            for key in [
                "document_id",
                "source_id",
                "char_count",
                "content_origin",
                "content_authority_type",
                "tags",
            ]
            if key in item
        }

    if source_type == "book":

        result = {
            key: item.get(key)
            for key in [
                "document_id",
                "work_id",
                "edition_id",
                "paragraph_number",
                "heading_depth",
                "heading_path",
                "matched_markers",
                "content_id",
                "parent_paragraph_id",
                "record_count",
                "review_status",
            ]
            if key in item
        }

        if "document_ids" in item:
            result[
                "document_id_count"
            ] = len(
                item["document_ids"]
            )

            result[
                "document_id_preview"
            ] = item[
                "document_ids"
            ][:3]

        return result

    if source_type == "transcript":
        return {
            key: item.get(key)
            for key in [
                "physical_file_id",
                "program_id",
                "episode_number",
                "season_number",
                "rows",
                "effective_labels",
                "primary_states",
                "left_document_id",
                "right_document_id",
                "left_utterance_index",
                "right_utterance_index",
                "left_effective_label",
                "right_effective_label",
                "left_primary",
                "right_primary",
                "review_status",
                "qa_relation_confirmed",
            ]
            if key in item
        }

    return item


# ============================================================
# Candidate containers
# ============================================================

candidate_pools = {
    "website":
        defaultdict(list),

    "book":
        defaultdict(list),

    "transcript":
        defaultdict(list),
}


candidate_states = {
    "website": {},
    "book": {},
    "transcript": {},
}


# ============================================================
# Website + Book discovery
# ============================================================

book_section_groups = defaultdict(
    list
)


with corpus_path.open(
    "r",
    encoding="utf-8",
) as f:

    for line in f:

        row = json.loads(line)

        source_type = row.get(
            "source_type"
        )

        text = str(
            row.get("text")
            or ""
        )

        attribution = (
            row.get("attribution")
            or {}
        )

        metadata = (
            row.get("source_metadata")
            or {}
        )

        locator = (
            row.get("locator")
            or {}
        )

        document_id = row.get(
            "document_id"
        )

        source_id = row.get(
            "source_id"
        )

        # ====================================================
        # WEBSITE
        # ====================================================

        if source_type == "website":

            content_origin = row.get(
                "content_origin"
            )

            authority_type = (
                attribution.get(
                    "content_authority_type"
                )
            )

            char_count = len(text)

            base_identity = {
                "document_id":
                    document_id,

                "source_id":
                    source_id,

                "char_count":
                    char_count,

                "content_origin":
                    content_origin,

                "content_authority_type":
                    authority_type,

                "canonical_url":
                    locator.get(
                        "canonical_url"
                    ),

                "url":
                    locator.get(
                        "url"
                    ),
            }

            if (
                content_origin
                == "primary_author"

                and authority_type
                == "direct_author"
            ):

                if (
                    char_count
                    <=
                    WEBSITE_LENGTH_BANDS_V1[
                        "direct_author"
                    ][
                        "short_max_chars"
                    ]
                ):
                    candidate_pools[
                        "website"
                    ][
                        "direct_author_short"
                    ].append(
                        dict(
                            base_identity
                        )
                    )

                if (
                    char_count
                    >=
                    WEBSITE_LENGTH_BANDS_V1[
                        "direct_author"
                    ][
                        "long_min_chars"
                    ]
                ):
                    candidate_pools[
                        "website"
                    ][
                        "direct_author_long"
                    ].append(
                        dict(
                            base_identity
                        )
                    )

            if (
                content_origin
                == "supervised"

                and authority_type
                == "editorial_supervision"
            ):

                if (
                    char_count
                    <=
                    WEBSITE_LENGTH_BANDS_V1[
                        "supervised"
                    ][
                        "short_max_chars"
                    ]
                ):
                    candidate_pools[
                        "website"
                    ][
                        "supervised_short"
                    ].append(
                        dict(
                            base_identity
                        )
                    )

                if (
                    char_count
                    >=
                    WEBSITE_LENGTH_BANDS_V1[
                        "supervised"
                    ][
                        "long_min_chars"
                    ]
                ):
                    candidate_pools[
                        "website"
                    ][
                        "supervised_long"
                    ].append(
                        dict(
                            base_identity
                        )
                    )

            tags = normalized_tags(
                row
            )

            if len(tags) >= 2:

                candidate_pools[
                    "website"
                ][
                    "multi_topic_article"
                ].append(
                    {
                        **base_identity,

                        "tags":
                            tags,

                        "review_status":
                            "HUMAN_REVIEW_REQUIRED",
                    }
                )

        # ====================================================
        # BOOK
        # ====================================================

        elif source_type == "book":

            heading_path = (
                locator.get(
                    "heading_path"
                )
                or
                metadata.get(
                    "heading_path"
                )
            )

            heading_depth = (
                heading_depth_from_value(
                    heading_path
                )
            )

            book_identity = {
                "document_id":
                    document_id,

                "source_id":
                    source_id,

                "work_id":
                    locator.get(
                        "work_id"
                    ),

                "edition_id":
                    locator.get(
                        "edition_id"
                    ),

                "paragraph_id":
                    locator.get(
                        "paragraph_id"
                    ),

                "paragraph_number":
                    locator.get(
                        "paragraph_number"
                    ),

                "heading_path":
                    heading_path,

                "heading_depth":
                    heading_depth,
            }

            activity_markers = (
                explicit_activity_marker(
                    row
                )
            )

            if activity_markers:

                candidate_pools[
                    "book"
                ][
                    "exercise_or_self_assessment"
                ].append(
                    {
                        **book_identity,

                        "matched_markers":
                            activity_markers,
                    }
                )

            if (
                locator.get(
                    "content_id"
                )
                is not None

                and

                locator.get(
                    "parent_paragraph_id"
                )
                is not None
            ):

                candidate_pools[
                    "book"
                ][
                    "supplemental_footnote_case"
                ].append(
                    {
                        **book_identity,

                        "content_id":
                            locator.get(
                                "content_id"
                            ),

                        "parent_paragraph_id":
                            locator.get(
                                "parent_paragraph_id"
                            ),

                        "source_locator":
                            locator.get(
                                "source_locator"
                            ),
                    }
                )

            if (
                heading_depth
                >= deep_heading_threshold
            ):

                candidate_pools[
                    "book"
                ][
                    "deep_heading_path"
                ].append(
                    dict(
                        book_identity
                    )
                )

            section_key = (
                locator.get(
                    "work_id"
                ),

                locator.get(
                    "edition_id"
                ),

                stable_json_key(
                    heading_path
                ),
            )

            book_section_groups[
                section_key
            ].append(
                {
                    **book_identity,

                    "is_heading":
                        metadata.get(
                            "is_heading"
                        ),

                    "in_table":
                        metadata.get(
                            "in_table"
                        ),
                }
            )


# ============================================================
# Book semantic-review prefilter
# ============================================================

for section_key, records in (
    book_section_groups.items()
):

    evidence_records = [
        record
        for record in records
        if (
            record.get(
                "is_heading"
            )
            is not True
        )
    ]

    distinct_documents = {
        record[
            "document_id"
        ]
        for record
        in evidence_records
    }

    if len(
        distinct_documents
    ) < 2:
        continue

    candidate_pools[
        "book"
    ][
        "multi_paragraph_evidence_case"
    ].append(
        {
            "work_id":
                section_key[0],

            "edition_id":
                section_key[1],

            "heading_path_key":
                section_key[2],

            "record_count":
                len(
                    evidence_records
                ),

            "document_ids":
                sorted(
                    distinct_documents
                ),

            "review_status":
                "HUMAN_REVIEW_REQUIRED",
        }
    )


# ============================================================
# Transcript file-level pools
# ============================================================

multi_program_map = {
    "program:uni-tehran":
        "multi_speaker_uni_tehran",

    "program:chehelston":
        "multi_speaker_chehelston",

    "program:asre-shirin":
        "multi_speaker_asre_shirin",

    "program:khane-to":
        "multi_speaker_khane_to",

    "program:mizan":
        "multi_speaker_mizan",
}


for profile in file_profiles:

    labels = profile[
        "labels"
    ]

    primary_states = profile[
        "primary_states"
    ]

    # --------------------------------------------------------
    # single_speaker_primary
    # --------------------------------------------------------

    if (
        len(labels) == 1

        and

        set(
            primary_states
        )
        == {"True"}
    ):

        candidate_pools[
            "transcript"
        ][
            "single_speaker_primary"
        ].append(
            {
                "physical_file_id":
                    profile[
                        "physical_file_id"
                    ],

                "program_id":
                    profile[
                        "program_id"
                    ],

                "episode_number":
                    profile[
                        "episode_number"
                    ],

                "season_number":
                    profile[
                        "season_number"
                    ],

                "rows":
                    profile[
                        "rows"
                    ],

                "effective_labels":
                    sorted(
                        scalarize(x)
                        for x
                        in labels
                    ),
            }
        )

    # --------------------------------------------------------
    # single_speaker_unresolved
    #
    # This implements the CURRENT rule exactly.
    # A zero match is preserved as evidence for rule review.
    # --------------------------------------------------------

    if (
        len(labels) == 1

        and

        profile[
            "unresolved_rows"
        ]
        == profile[
            "rows"
        ]
    ):

        candidate_pools[
            "transcript"
        ][
            "single_speaker_unresolved"
        ].append(
            {
                "physical_file_id":
                    profile[
                        "physical_file_id"
                    ],

                "program_id":
                    profile[
                        "program_id"
                    ],

                "episode_number":
                    profile[
                        "episode_number"
                    ],

                "season_number":
                    profile[
                        "season_number"
                    ],

                "rows":
                    profile[
                        "rows"
                    ],

                "unknown_speaker_inference":
                    "FORBIDDEN",
            }
        )

    # --------------------------------------------------------
    # required multi-speaker programs
    # --------------------------------------------------------

    target_stratum = (
        multi_program_map.get(
            profile[
                "program_id"
            ]
        )
    )

    if (
        target_stratum
        is not None

        and

        len(labels) > 1
    ):

        candidate_pools[
            "transcript"
        ][
            target_stratum
        ].append(
            {
                "physical_file_id":
                    profile[
                        "physical_file_id"
                    ],

                "program_id":
                    profile[
                        "program_id"
                    ],

                "episode_number":
                    profile[
                        "episode_number"
                    ],

                "season_number":
                    profile[
                        "season_number"
                    ],

                "rows":
                    profile[
                        "rows"
                    ],

                "effective_labels":
                    sorted(
                        scalarize(x)
                        for x
                        in labels
                    ),
            }
        )

    # --------------------------------------------------------
    # mixed_primary_signal_case
    # --------------------------------------------------------

    if len(
        primary_states
    ) > 1:

        candidate_pools[
            "transcript"
        ][
            "mixed_primary_signal_case"
        ].append(
            {
                "physical_file_id":
                    profile[
                        "physical_file_id"
                    ],

                "program_id":
                    profile[
                        "program_id"
                    ],

                "episode_number":
                    profile[
                        "episode_number"
                    ],

                "season_number":
                    profile[
                        "season_number"
                    ],

                "rows":
                    profile[
                        "rows"
                    ],

                "primary_states":
                    dict(
                        primary_states
                    ),
            }
        )


# ============================================================
# Transcript adjacency pools
# ============================================================

for physical_file_id, rows in (
    transcript_files.items()
):

    for left, right in zip(
        rows,
        rows[1:],
    ):

        # ----------------------------------------------------
        # primary_to_nonprimary_transition
        #
        # Exact current rule:
        # True -> False only.
        #
        # None/Unresolved is NOT silently relabeled as False.
        # Zero matches remain evidence for next rule audit.
        # ----------------------------------------------------

        if (
            left[
                "primary_dr_roudi_corpus"
            ]
            is True

            and

            right[
                "primary_dr_roudi_corpus"
            ]
            is False
        ):

            candidate_pools[
                "transcript"
            ][
                "primary_to_nonprimary_transition"
            ].append(
                {
                    "physical_file_id":
                        physical_file_id,

                    "program_id":
                        left[
                            "program_id"
                        ],

                    "left_document_id":
                        left[
                            "document_id"
                        ],

                    "right_document_id":
                        right[
                            "document_id"
                        ],

                    "left_utterance_index":
                        left[
                            "utterance_index"
                        ],

                    "right_utterance_index":
                        right[
                            "utterance_index"
                        ],

                    "left_primary":
                        True,

                    "right_primary":
                        False,
                }
            )

        # ----------------------------------------------------
        # Question/answer REVIEW PREFILTER only
        # ----------------------------------------------------

        left_label = left.get(
            "effective_label"
        )

        right_label = right.get(
            "effective_label"
        )

        if (
            left_label
            is not None

            and

            right_label
            is not None

            and

            left_label
            != right_label
        ):

            candidate_pools[
                "transcript"
            ][
                "question_answer_transition"
            ].append(
                {
                    "physical_file_id":
                        physical_file_id,

                    "program_id":
                        left[
                            "program_id"
                        ],

                    "left_document_id":
                        left[
                            "document_id"
                        ],

                    "right_document_id":
                        right[
                            "document_id"
                        ],

                    "left_utterance_index":
                        left[
                            "utterance_index"
                        ],

                    "right_utterance_index":
                        right[
                            "utterance_index"
                        ],

                    "left_effective_label":
                        left_label,

                    "right_effective_label":
                        right_label,

                    "review_status":
                        "HUMAN_REVIEW_REQUIRED",

                    "qa_relation_confirmed":
                        False,
                }
            )


# ============================================================
# Discovery classification
# ============================================================

exact_rule_strata = {
    "website": {
        "direct_author_short",
        "direct_author_long",
        "supervised_short",
        "supervised_long",
    },

    "book": {
        "exercise_or_self_assessment",
        "supplemental_footnote_case",
        "deep_heading_path",
    },

    "transcript": {
        "single_speaker_primary",
        "single_speaker_unresolved",
        "multi_speaker_uni_tehran",
        "multi_speaker_chehelston",
        "multi_speaker_asre_shirin",
        "multi_speaker_khane_to",
        "multi_speaker_mizan",
        "primary_to_nonprimary_transition",
        "mixed_primary_signal_case",
    },
}


review_prefilter_strata = {
    "website": {
        "multi_topic_article",
    },

    "book": {
        "multi_paragraph_evidence_case",
    },

    "transcript": {
        "question_answer_transition",
    },
}


dependency_required = {

    "website": {
        "citation_locator_sensitive_case":
            "NB09_DUPLICATE_OR_SIMILARITY_LINKAGE_DISCOVERY_REQUIRED",
    },

    "book": {
        "intrinsically_tiny_narrative_region":
            "EXACT_PROTECTED_REGION_TOKEN_IDENTITY_REQUIRED",

        "intrinsically_tiny_table_region":
            "EXACT_PROTECTED_TABLE_REGION_TOKEN_IDENTITY_REQUIRED",

        "long_section":
            "EXACT_BOOK_SECTION_TOKEN_MEASUREMENT_REQUIRED",

        "large_table_run":
            "EXACT_TABLE_RUN_TOKEN_MEASUREMENT_REQUIRED",
    },

    "transcript": {
        "long_utterance_over_500":
            "EXACT_SOURCE_UTTERANCE_TOKEN_MEASUREMENT_REQUIRED",

        "long_utterance_over_800":
            "EXACT_SOURCE_UTTERANCE_TOKEN_MEASUREMENT_REQUIRED",

        "token_fallback_sentence_case":
            "EXACT_SENTENCE_TOKEN_AND_FALLBACK_LINKAGE_REQUIRED",
    },
}


zero_match_exact_rules = []


for source_type in [
    "website",
    "book",
    "transcript",
]:

    for stratum in expected_strata[
        source_type
    ]:

        if (
            stratum
            in exact_rule_strata[
                source_type
            ]
        ):

            count = len(
                candidate_pools[
                    source_type
                ][
                    stratum
                ]
            )

            if count > 0:

                status = (
                    "EXACT_POOL_READY"
                )

            else:

                status = (
                    "EXACT_RULE_ZERO_MATCH_REVIEW_REQUIRED"
                )

                zero_match_exact_rules.append(
                    {
                        "source_type":
                            source_type,

                        "stratum":
                            stratum,

                        "candidate_count":
                            0,

                        "interpretation":
                            (
                                "ZERO_MATCH_UNDER_CURRENT_RULE; "
                                "DO_NOT_FABRICATE; "
                                "REVIEW_RULE_AGAINST_REAL_SCHEMA"
                            ),
                    }
                )

            candidate_states[
                source_type
            ][
                stratum
            ] = {
                "status":
                    status,

                "candidate_count":
                    count,
            }

        elif (
            stratum
            in review_prefilter_strata[
                source_type
            ]
        ):

            candidate_states[
                source_type
            ][
                stratum
            ] = {
                "status":
                    "REVIEW_PREFILTER_READY",

                "candidate_count":
                    len(
                        candidate_pools[
                            source_type
                        ][
                            stratum
                        ]
                    ),

                "final_membership_confirmed":
                    False,
            }

        elif (
            stratum
            in dependency_required[
                source_type
            ]
        ):

            candidate_states[
                source_type
            ][
                stratum
            ] = {
                "status":
                    "DEPENDENCY_REQUIRED",

                "candidate_count":
                    None,

                "dependency":
                    dependency_required[
                        source_type
                    ][
                        stratum
                    ],
            }

        else:

            candidate_states[
                source_type
            ][
                stratum
            ] = {
                "status":
                    "UNCLASSIFIED",
            }


# ============================================================
# Validation
# ============================================================

all_27_classified_gate = (
    sum(
        len(
            source_states
        )
        for source_states
        in candidate_states.values()
    )
    == 27
)


no_unclassified_gate = all(
    state[
        "status"
    ]
    != "UNCLASSIFIED"

    for source_states
    in candidate_states.values()

    for state
    in source_states.values()
)


all_evaluated_pool_counts_valid_gate = all(
    isinstance(
        state.get(
            "candidate_count"
        ),
        int,
    )

    and

    state[
        "candidate_count"
    ]
    >= 0

    for source_type, source_states
    in candidate_states.items()

    for stratum, state
    in source_states.items()

    if state[
        "status"
    ]
    in {
        "EXACT_POOL_READY",
        "EXACT_RULE_ZERO_MATCH_REVIEW_REQUIRED",
        "REVIEW_PREFILTER_READY",
    }
)


zero_match_rules_explicit_gate = all(
    item[
        "candidate_count"
    ]
    == 0

    and

    candidate_states[
        item[
            "source_type"
        ]
    ][
        item[
            "stratum"
        ]
    ][
        "status"
    ]
    ==
    "EXACT_RULE_ZERO_MATCH_REVIEW_REQUIRED"

    for item
    in zero_match_exact_rules
)


dependency_counts_not_fabricated_gate = all(
    candidate_states[
        source_type
    ][
        stratum
    ][
        "candidate_count"
    ]
    is None

    for source_type, dependency_map
    in dependency_required.items()

    for stratum
    in dependency_map
)


qa_prefilter_not_confirmed_gate = all(
    item[
        "qa_relation_confirmed"
    ]
    is False

    for item
    in candidate_pools[
        "transcript"
    ][
        "question_answer_transition"
    ]
)


unknown_speaker_not_inferred_gate = (
    all(
        item.get(
            "unknown_speaker_inference"
        )
        == "FORBIDDEN"

        for item
        in candidate_pools[
            "transcript"
        ][
            "single_speaker_unresolved"
        ]
    )
)


sampling_not_started_gate = (
    GOLDEN_DATASET_CONTRACT_V1_DRAFT[
        "sampling_policy"
    ][
        "status"
    ]
    == "NOT_YET_DEFINED"
)


question_generation_not_started_gate = (
    GOLDEN_DATASET_CONTRACT_V1_DRAFT[
        "question_answer_policy"
    ][
        "question_generation_status"
    ]
    == "NOT_STARTED"
)


candidate_discovery_gates = {

    "all_27_strata_classified":
        all_27_classified_gate,

    "no_unclassified_strata":
        no_unclassified_gate,

    "evaluated_pool_counts_are_valid":
        all_evaluated_pool_counts_valid_gate,

    "zero_match_exact_rules_explicitly_flagged":
        zero_match_rules_explicit_gate,

    "unmeasured_dependencies_not_fabricated":
        dependency_counts_not_fabricated_gate,

    "qa_prefilter_not_mislabeled_as_confirmed":
        qa_prefilter_not_confirmed_gate,

    "unknown_speaker_not_inferred":
        unknown_speaker_not_inferred_gate,

    "sampling_not_started":
        sampling_not_started_gate,

    "question_generation_not_started":
        question_generation_not_started_gate,
}


candidate_discovery_pass = all(
    candidate_discovery_gates.values()
)


# ============================================================
# Output
# ============================================================

print(
    "=== CANDIDATE DISCOVERY — STAGE A (REPAIRED) ==="
)


status_totals = Counter()


for source_type in [
    "website",
    "book",
    "transcript",
]:

    print(
        f"\n[{source_type.upper()}]"
    )

    for stratum in expected_strata[
        source_type
    ]:

        state = candidate_states[
            source_type
        ][
            stratum
        ]

        status_totals[
            state[
                "status"
            ]
        ] += 1

        print(
            "-",
            stratum,
            "| status:",
            state[
                "status"
            ],
            "| candidates:",
            state.get(
                "candidate_count"
            ),
        )

        if state.get(
            "dependency"
        ):

            print(
                "  dependency:",
                state[
                    "dependency"
                ]
            )


print(
    "\n=== ZERO-MATCH EXACT RULES ==="
)

if zero_match_exact_rules:

    for item in (
        zero_match_exact_rules
    ):

        print(
            "-",
            item[
                "source_type"
            ],
            "/",
            item[
                "stratum"
            ],
            "| candidates:",
            item[
                "candidate_count"
            ],
            "| status: RULE_REVIEW_REQUIRED",
        )

else:

    print(
        "NONE"
    )


print(
    "\n=== STATUS TOTALS ==="
)

for status, count in sorted(
    status_totals.items()
):

    print(
        "-",
        status,
        ":",
        count,
    )


print(
    "\n=== COMPACT REPRESENTATIVE IDENTITIES ==="
)

for source_type in [
    "website",
    "book",
    "transcript",
]:

    print(
        f"\n[{source_type.upper()}]"
    )

    for stratum in expected_strata[
        source_type
    ]:

        pool = candidate_pools[
            source_type
        ][
            stratum
        ]

        if not pool:
            continue

        print(
            "\n",
            stratum,
            "| examples:",
        )

        for example in pool[:2]:

            print(
                json.dumps(
                    compact_example(
                        source_type,
                        stratum,
                        example,
                    ),
                    ensure_ascii=False,
                    sort_keys=True,
                )
            )


print(
    "\n=== CANDIDATE DISCOVERY GATES ==="
)

for name, passed in (
    candidate_discovery_gates.items()
):

    print(
        name,
        ":",
        "PASS"
        if passed
        else "FAIL"
    )


print(
    "\nCODE CELL 09 GATE:",
    "PASS"
    if candidate_discovery_pass
    else "FAIL"
)


assert candidate_discovery_pass, (
    "Candidate Discovery Stage A validation failed. "
    "Do not proceed until classification, zero-match handling, "
    "and dependency accounting are consistent."
)
```

<div style="direction: rtl; white-space: normal; line-height: 1.7;">

<h3>Zero-Match Transcript Rule Audit</h3>

<p>
Candidate Discovery Stage A با موفقیت PASS شد، اما دو Stratum تحت Rule فعلی هیچ Candidateای تولید نکردند:
<code>single_speaker_unresolved</code> و <code>primary_to_nonprimary_transition</code>.
</p>

<p>
Zero Match به‌تنهایی به معنی نبود پدیده در Corpus نیست.
ممکن است Operational Rule با Representation واقعی Metadata هم‌راستا نباشد.
بنابراین قبل از تغییر Rule، باید Evidence واقعی Transcript بررسی شود.
</p>

<p>
برای <code>single_speaker_unresolved</code> بررسی می‌شود که Records با
<code>base_status = UNRESOLVED</code> چگونه در Physical Fileها توزیع شده‌اند،
آیا Effective Speaker Label دارند یا ندارند و آیا File کاملاً unresolved واقعاً وجود دارد.
</p>

<p>
برای <code>primary_to_nonprimary_transition</code> ابتدا مقادیر واقعی
<code>primary_dr_roudi_corpus</code> بررسی می‌شوند.
سپس Transitionهای <code>True → False</code> و <code>True → None</code>
به‌صورت جداگانه شمارش می‌شوند.
مقدار <code>None</code> بدون Evidence صریح هرگز به <code>False</code> تبدیل نخواهد شد.
</p>

<p>
این مرحله فقط Audit است.
هیچ Ruleای در این Cell تغییر نمی‌کند و هیچ Candidate جدیدی پذیرفته نمی‌شود.
</p>

</div>

```python
# Code Cell 10

# ============================================================
# ZERO-MATCH AUDIT:
# 1) single_speaker_unresolved
# 2) primary_to_nonprimary_transition
#
# No rule mutation in this cell.
# ============================================================


# ============================================================
# Row-level transcript state inventory
# ============================================================

primary_signal_counts = Counter()
base_status_counts = Counter()

false_primary_rows = []
unresolved_rows_all = []

file_audit = {}


for physical_file_id, rows in (
    transcript_files.items()
):

    row_primary_states = Counter()
    row_base_statuses = Counter()

    effective_labels = set()
    effective_canonicals = set()
    raw_labels = set()

    unresolved_rows = []

    for row in rows:

        primary_value = row.get(
            "primary_dr_roudi_corpus"
        )

        base_status = row.get(
            "base_status"
        )

        effective_label = row.get(
            "effective_label"
        )

        effective_canonical = row.get(
            "effective_canonical"
        )

        base_raw = row.get(
            "base_raw"
        )

        primary_signal_counts[
            scalarize(
                primary_value
            )
        ] += 1

        base_status_counts[
            scalarize(
                base_status
            )
        ] += 1

        row_primary_states[
            scalarize(
                primary_value
            )
        ] += 1

        row_base_statuses[
            scalarize(
                base_status
            )
        ] += 1

        if effective_label is not None:
            effective_labels.add(
                effective_label
            )

        if effective_canonical is not None:
            effective_canonicals.add(
                effective_canonical
            )

        if base_raw is not None:
            raw_labels.add(
                base_raw
            )

        if primary_value is False:
            false_primary_rows.append(
                {
                    "physical_file_id":
                        physical_file_id,

                    "program_id":
                        row.get(
                            "program_id"
                        ),

                    "document_id":
                        row.get(
                            "document_id"
                        ),

                    "utterance_index":
                        row.get(
                            "utterance_index"
                        ),

                    "base_status":
                        base_status,

                    "base_raw":
                        base_raw,

                    "effective_label":
                        effective_label,

                    "effective_canonical":
                        effective_canonical,
                }
            )

        if base_status == "UNRESOLVED":

            unresolved_record = {
                "physical_file_id":
                    physical_file_id,

                "program_id":
                    row.get(
                        "program_id"
                    ),

                "document_id":
                    row.get(
                        "document_id"
                    ),

                "utterance_index":
                    row.get(
                        "utterance_index"
                    ),

                "base_raw":
                    base_raw,

                "effective_label":
                    effective_label,

                "effective_canonical":
                    effective_canonical,

                "primary_dr_roudi_corpus":
                    primary_value,
            }

            unresolved_rows.append(
                unresolved_record
            )

            unresolved_rows_all.append(
                unresolved_record
            )

    file_audit[
        physical_file_id
    ] = {
        "physical_file_id":
            physical_file_id,

        "program_id":
            rows[0].get(
                "program_id"
            )
            if rows
            else None,

        "episode_number":
            rows[0].get(
                "episode_number"
            )
            if rows
            else None,

        "season_number":
            rows[0].get(
                "season_number"
            )
            if rows
            else None,

        "rows":
            len(rows),

        "primary_states":
            dict(
                row_primary_states
            ),

        "base_statuses":
            dict(
                row_base_statuses
            ),

        "effective_labels":
            sorted(
                scalarize(value)
                for value
                in effective_labels
            ),

        "effective_canonicals":
            sorted(
                scalarize(value)
                for value
                in effective_canonicals
            ),

        "raw_labels":
            sorted(
                scalarize(value)
                for value
                in raw_labels
            ),

        "unresolved_rows":
            unresolved_rows,
    }


# ============================================================
# Audit 1: single_speaker_unresolved
# ============================================================

files_with_unresolved = [
    profile
    for profile
    in file_audit.values()
    if profile[
        "unresolved_rows"
    ]
]


files_all_rows_unresolved = [
    profile
    for profile
    in files_with_unresolved
    if len(
        profile[
            "unresolved_rows"
        ]
    )
    == profile[
        "rows"
    ]
]


all_unresolved_by_effective_label_count = Counter()

all_unresolved_by_raw_label_count = Counter()

all_unresolved_by_canonical_count = Counter()


for profile in (
    files_all_rows_unresolved
):

    all_unresolved_by_effective_label_count[
        len(
            profile[
                "effective_labels"
            ]
        )
    ] += 1

    all_unresolved_by_raw_label_count[
        len(
            profile[
                "raw_labels"
            ]
        )
    ] += 1

    all_unresolved_by_canonical_count[
        len(
            profile[
                "effective_canonicals"
            ]
        )
    ] += 1


unresolved_rows_by_program = Counter(
    item[
        "program_id"
    ]
    for item
    in unresolved_rows_all
)


# ============================================================
# Audit 2: primary_to_nonprimary_transition
# ============================================================

adjacent_primary_state_transitions = Counter()

true_to_none_examples = []

true_to_false_examples = []

none_to_true_examples = []


for physical_file_id, rows in (
    transcript_files.items()
):

    for left, right in zip(
        rows,
        rows[1:],
    ):

        left_state = left.get(
            "primary_dr_roudi_corpus"
        )

        right_state = right.get(
            "primary_dr_roudi_corpus"
        )

        transition_key = (
            scalarize(left_state),
            scalarize(right_state),
        )

        adjacent_primary_state_transitions[
            transition_key
        ] += 1

        common = {
            "physical_file_id":
                physical_file_id,

            "program_id":
                left.get(
                    "program_id"
                ),

            "left_document_id":
                left.get(
                    "document_id"
                ),

            "right_document_id":
                right.get(
                    "document_id"
                ),

            "left_utterance_index":
                left.get(
                    "utterance_index"
                ),

            "right_utterance_index":
                right.get(
                    "utterance_index"
                ),

            "left_base_status":
                left.get(
                    "base_status"
                ),

            "right_base_status":
                right.get(
                    "base_status"
                ),

            "left_effective_label":
                left.get(
                    "effective_label"
                ),

            "right_effective_label":
                right.get(
                    "effective_label"
                ),

            "left_effective_canonical":
                left.get(
                    "effective_canonical"
                ),

            "right_effective_canonical":
                right.get(
                    "effective_canonical"
                ),

            "left_primary":
                left_state,

            "right_primary":
                right_state,
        }

        if (
            left_state is True
            and right_state is None
            and len(
                true_to_none_examples
            ) < 20
        ):
            true_to_none_examples.append(
                common
            )

        if (
            left_state is True
            and right_state is False
            and len(
                true_to_false_examples
            ) < 20
        ):
            true_to_false_examples.append(
                common
            )

        if (
            left_state is None
            and right_state is True
            and len(
                none_to_true_examples
            ) < 20
        ):
            none_to_true_examples.append(
                common
            )


# ============================================================
# Output
# ============================================================

print(
    "=== TRANSCRIPT PRIMARY SIGNAL INVENTORY ==="
)

print(
    "primary_dr_roudi_corpus:",
    primary_signal_counts
)

print(
    "base_status:",
    base_status_counts
)

print(
    "rows with explicit primary=False:",
    len(
        false_primary_rows
    )
)


if false_primary_rows:

    print(
        "\nExplicit False examples:"
    )

    for item in (
        false_primary_rows[:20]
    ):
        print(
            json.dumps(
                item,
                ensure_ascii=False,
                sort_keys=True,
            )
        )


print(
    "\n=== ZERO RULE AUDIT 1: single_speaker_unresolved ==="
)

print(
    "UNRESOLVED rows:",
    len(
        unresolved_rows_all
    )
)

print(
    "physical files containing >=1 UNRESOLVED row:",
    len(
        files_with_unresolved
    )
)

print(
    "physical files with ALL rows UNRESOLVED:",
    len(
        files_all_rows_unresolved
    )
)

print(
    "all-unresolved files by effective-label count:",
    all_unresolved_by_effective_label_count
)

print(
    "all-unresolved files by raw-label count:",
    all_unresolved_by_raw_label_count
)

print(
    "all-unresolved files by canonical count:",
    all_unresolved_by_canonical_count
)

print(
    "\nUNRESOLVED rows by program:"
)

for program_id, count in (
    unresolved_rows_by_program.most_common()
):
    print(
        "-",
        program_id,
        ":",
        count
    )


print(
    "\nRepresentative all-unresolved file profiles:"
)

for profile in (
    files_all_rows_unresolved[:10]
):

    printable = {
        "physical_file_id":
            profile[
                "physical_file_id"
            ],

        "program_id":
            profile[
                "program_id"
            ],

        "episode_number":
            profile[
                "episode_number"
            ],

        "season_number":
            profile[
                "season_number"
            ],

        "rows":
            profile[
                "rows"
            ],

        "primary_states":
            profile[
                "primary_states"
            ],

        "base_statuses":
            profile[
                "base_statuses"
            ],

        "effective_labels":
            profile[
                "effective_labels"
            ],

        "effective_canonicals":
            profile[
                "effective_canonicals"
            ],

        "raw_labels":
            profile[
                "raw_labels"
            ],
    }

    print(
        json.dumps(
            printable,
            ensure_ascii=False,
            sort_keys=True,
        )
    )


print(
    "\n=== ZERO RULE AUDIT 2: primary_to_nonprimary_transition ==="
)

print(
    "Adjacent primary-state transitions:"
)

for (
    left_state,
    right_state
), count in sorted(
    adjacent_primary_state_transitions.items()
):

    print(
        "-",
        f"{left_state} -> {right_state}",
        ":",
        count
    )


print(
    "\nTrue -> False count:",
    adjacent_primary_state_transitions[
        (
            "True",
            "False",
        )
    ]
)

print(
    "True -> None count:",
    adjacent_primary_state_transitions[
        (
            "True",
            "None",
        )
    ]
)

print(
    "None -> True count:",
    adjacent_primary_state_transitions[
        (
            "None",
            "True",
        )
    ]
)


print(
    "\nRepresentative True -> False transitions:"
)

if true_to_false_examples:

    for item in (
        true_to_false_examples
    ):
        print(
            json.dumps(
                item,
                ensure_ascii=False,
                sort_keys=True,
            )
        )

else:

    print(
        "NONE"
    )


print(
    "\nRepresentative True -> None transitions:"
)

for item in (
    true_to_none_examples[:10]
):
    print(
        json.dumps(
            item,
            ensure_ascii=False,
            sort_keys=True,
        )
    )


print(
    "\nRepresentative None -> True transitions:"
)

for item in (
    none_to_true_examples[:10]
):
    print(
        json.dumps(
            item,
            ensure_ascii=False,
            sort_keys=True,
        )
    )


# ============================================================
# Audit flags — descriptive, not rule mutation
# ============================================================

zero_rule_audit_flags = {

    "unresolved_rows_exist":
        len(
            unresolved_rows_all
        )
        > 0,

    "all_unresolved_files_exist":
        len(
            files_all_rows_unresolved
        )
        > 0,

    "explicit_false_primary_rows_exist":
        len(
            false_primary_rows
        )
        > 0,

    "true_to_false_transitions_exist":
        adjacent_primary_state_transitions[
            (
                "True",
                "False",
            )
        ]
        > 0,

    "true_to_none_transitions_exist":
        adjacent_primary_state_transitions[
            (
                "True",
                "None",
            )
        ]
        > 0,

    "none_to_true_transitions_exist":
        adjacent_primary_state_transitions[
            (
                "None",
                "True",
            )
        ]
        > 0,
}


print(
    "\n=== ZERO-RULE AUDIT FLAGS ==="
)

for name, value in (
    zero_rule_audit_flags.items()
):

    print(
        name,
        ":",
        "YES" if value else "NO"
    )


audit_complete = (
    len(
        unresolved_rows_all
    )
    > 0

    and

    sum(
        adjacent_primary_state_transitions.values()
    )
    > 0
)


print(
    "\nCODE CELL 10 GATE:",
    "PASS"
    if audit_complete
    else "FAIL"
)


assert audit_complete, (
    "Zero-match rule audit did not recover enough "
    "real transcript evidence for diagnosis."
)
```

<h3>Repair Single-Speaker Unresolved Rule and Audit Explicit Non-primary Transitions</h3>

<p>
Zero-match Audit نشان داد که Rule قبلی برای
<code>single_speaker_unresolved</code> با Representation واقعی Corpus هم‌راستا نبود.
هفت Physical File وجود دارند که تمام Records آن‌ها
<code>base_status = UNRESOLVED</code> دارند، اما هیچ Effective Speaker Label یا Canonical Identity ندارند.
</p>

<p>
بنابراین تعریف صحیح این Stratum نباید وجود یک Speaker Label را الزام کند.
برعکس، Case معتبر باید تماماً Unresolved باشد و هیچ هویت مؤثری برای Speaker به آن نسبت داده نشده باشد.
این اصلاح بدون هیچ Speaker Inference انجام می‌شود.
</p>

<p>
برای <code>primary_to_nonprimary_transition</code> هنوز Rule تغییر نمی‌کند.
Audit قبلی نشان داد <code>True → False</code> وجود ندارد، در حالی که
<code>True → None</code> فراوان است.
اما مقدار <code>None</code> هم Context صریح Non-primary و هم Speaker حل‌نشده را شامل می‌شود.
</p>

<p>
در این مرحله، Transitionهای <code>True → None</code> بر اساس Evidence واقعی
به گروه‌های مجزا تقسیم می‌شوند:
Explicit labeled context، Unresolved context و سایر وضعیت‌ها.
فقط پس از این تفکیک می‌توان تصمیم گرفت آیا Stratum
<code>primary_to_nonprimary_transition</code> Rule عملیاتی قابل دفاع دارد یا خیر.
</p>

<p>
هیچ <code>None</code> به‌صورت خودکار به Non-primary تبدیل نمی‌شود.
</p>

</div>

```python
# Code Cell 11

# ============================================================
# PART A
# Repair single_speaker_unresolved based on real evidence.
# ============================================================

STRATUM_RULES_V1_DRAFT[
    "transcript"
][
    "single_speaker_unresolved"
] = {
    "eligibility":
        "AUTO_ELIGIBLE",

    "unit":
        "PHYSICAL_FILE",

    "conditions": {
        "source_type":
            "transcript",

        "all_base_status":
            "UNRESOLVED",

        "effective_speaker_label_count":
            0,

        "effective_canonical_identity_count":
            0,
    },

    "unknown_speaker_inference":
        "FORBIDDEN",

    "interpretation": (
        "PHYSICAL_FILE_IS_ENTIRELY_UNRESOLVED_AND_NO_"
        "EFFECTIVE_SPEAKER_IDENTITY_HAS_BEEN_ASSIGNED"
    ),

    "evidence_basis":
        "CODE_CELL_10_ZERO_MATCH_AUDIT",
}


# Rebuild exact candidate pool under repaired rule.

candidate_pools[
    "transcript"
][
    "single_speaker_unresolved"
] = []


for profile in file_audit.values():

    all_rows_unresolved = (
        profile[
            "rows"
        ]
        > 0

        and

        profile[
            "base_statuses"
        ]
        == {
            "UNRESOLVED":
                profile[
                    "rows"
                ]
        }
    )

    no_effective_labels = (
        len(
            profile[
                "effective_labels"
            ]
        )
        == 0
    )

    no_effective_canonical = (
        len(
            profile[
                "effective_canonicals"
            ]
        )
        == 0
    )

    if (
        all_rows_unresolved
        and no_effective_labels
        and no_effective_canonical
    ):

        candidate_pools[
            "transcript"
        ][
            "single_speaker_unresolved"
        ].append(
            {
                "physical_file_id":
                    profile[
                        "physical_file_id"
                    ],

                "program_id":
                    profile[
                        "program_id"
                    ],

                "episode_number":
                    profile[
                        "episode_number"
                    ],

                "season_number":
                    profile[
                        "season_number"
                    ],

                "rows":
                    profile[
                        "rows"
                    ],

                "base_status":
                    "UNRESOLVED",

                "effective_speaker_count":
                    0,

                "effective_canonical_count":
                    0,

                "unknown_speaker_inference":
                    "FORBIDDEN",
            }
        )


candidate_states[
    "transcript"
][
    "single_speaker_unresolved"
] = {
    "status":
        "EXACT_POOL_READY",

    "candidate_count":
        len(
            candidate_pools[
                "transcript"
            ][
                "single_speaker_unresolved"
            ]
        ),

    "rule_status":
        "REPAIRED_FROM_CODE_CELL_10_EVIDENCE",
}


# ============================================================
# PART B
# Audit True -> None transitions without relabeling None.
# ============================================================

true_to_none_class_counts = Counter()

true_to_none_by_program = defaultdict(
    Counter
)

true_to_none_class_examples = defaultdict(
    list
)


for physical_file_id, rows in (
    transcript_files.items()
):

    for left, right in zip(
        rows,
        rows[1:],
    ):

        if not (
            left.get(
                "primary_dr_roudi_corpus"
            )
            is True

            and

            right.get(
                "primary_dr_roudi_corpus"
            )
            is None
        ):
            continue

        right_status = right.get(
            "base_status"
        )

        right_label = right.get(
            "effective_label"
        )

        right_canonical = right.get(
            "effective_canonical"
        )

        right_raw = right.get(
            "base_raw"
        )

        # ----------------------------------------------------
        # Classification is evidence-descriptive only.
        # No authority state is mutated here.
        # ----------------------------------------------------

        if right_status == "UNRESOLVED":

            transition_class = (
                "UNRESOLVED_CONTEXT"
            )

        elif (
            right_label is not None

            or

            right_canonical is not None

            or

            right_raw is not None
        ):

            transition_class = (
                "EXPLICIT_LABELED_CONTEXT"
            )

        else:

            transition_class = (
                "OTHER_NULL_PRIMARY_CONTEXT"
            )

        true_to_none_class_counts[
            transition_class
        ] += 1

        true_to_none_by_program[
            left.get(
                "program_id"
            )
        ][
            transition_class
        ] += 1

        if (
            len(
                true_to_none_class_examples[
                    transition_class
                ]
            )
            < 15
        ):

            true_to_none_class_examples[
                transition_class
            ].append(
                {
                    "physical_file_id":
                        physical_file_id,

                    "program_id":
                        left.get(
                            "program_id"
                        ),

                    "left_document_id":
                        left.get(
                            "document_id"
                        ),

                    "right_document_id":
                        right.get(
                            "document_id"
                        ),

                    "left_utterance_index":
                        left.get(
                            "utterance_index"
                        ),

                    "right_utterance_index":
                        right.get(
                            "utterance_index"
                        ),

                    "left_primary":
                        True,

                    "left_effective_label":
                        left.get(
                            "effective_label"
                        ),

                    "left_effective_canonical":
                        left.get(
                            "effective_canonical"
                        ),

                    "right_primary":
                        None,

                    "right_base_status":
                        right_status,

                    "right_base_raw":
                        right_raw,

                    "right_effective_label":
                        right_label,

                    "right_effective_canonical":
                        right_canonical,

                    "audit_class":
                        transition_class,
                }
            )


# ============================================================
# Explicit-labeled context deeper inventory
# ============================================================

explicit_context_label_counts = Counter()
explicit_context_status_counts = Counter()
explicit_context_canonical_counts = Counter()


for example in (
    true_to_none_class_examples[
        "EXPLICIT_LABELED_CONTEXT"
    ]
):

    explicit_context_label_counts[
        scalarize(
            example[
                "right_effective_label"
            ]
        )
    ] += 1

    explicit_context_status_counts[
        scalarize(
            example[
                "right_base_status"
            ]
        )
    ] += 1

    explicit_context_canonical_counts[
        scalarize(
            example[
                "right_effective_canonical"
            ]
        )
    ] += 1


# ============================================================
# Validation
# ============================================================

repaired_unresolved_count = len(
    candidate_pools[
        "transcript"
    ][
        "single_speaker_unresolved"
    ]
)


repaired_rule_matches_audit_gate = (
    repaired_unresolved_count
    == len(
        files_all_rows_unresolved
    )
    == 7
)


no_unresolved_identity_inference_gate = all(
    item[
        "effective_speaker_count"
    ]
    == 0

    and

    item[
        "effective_canonical_count"
    ]
    == 0

    and

    item[
        "unknown_speaker_inference"
    ]
    == "FORBIDDEN"

    for item
    in candidate_pools[
        "transcript"
    ][
        "single_speaker_unresolved"
    ]
)


true_to_none_accounting_gate = (
    sum(
        true_to_none_class_counts.values()
    )
    ==
    adjacent_primary_state_transitions[
        (
            "True",
            "None",
        )
    ]
)


no_none_relabel_gate = (
    candidate_states[
        "transcript"
    ][
        "primary_to_nonprimary_transition"
    ][
        "candidate_count"
    ]
    == 0

    and

    candidate_states[
        "transcript"
    ][
        "primary_to_nonprimary_transition"
    ][
        "status"
    ]
    ==
    "EXACT_RULE_ZERO_MATCH_REVIEW_REQUIRED"
)


transition_audit_has_evidence_gate = (
    true_to_none_class_counts[
        "EXPLICIT_LABELED_CONTEXT"
    ]
    > 0

    or

    true_to_none_class_counts[
        "UNRESOLVED_CONTEXT"
    ]
    > 0
)


cell_11_gates = {
    "single_speaker_unresolved_rule_repaired":
        repaired_rule_matches_audit_gate,

    "unknown_speaker_identity_not_inferred":
        no_unresolved_identity_inference_gate,

    "true_to_none_fully_accounted":
        true_to_none_accounting_gate,

    "none_not_reclassified_as_nonprimary":
        no_none_relabel_gate,

    "transition_audit_contains_real_evidence":
        transition_audit_has_evidence_gate,
}


cell_11_pass = all(
    cell_11_gates.values()
)


# ============================================================
# Output
# ============================================================

print(
    "=== REPAIRED STRATUM: single_speaker_unresolved ==="
)

print(
    "Candidate count:",
    repaired_unresolved_count
)

print(
    "Rule:"
)

print(
    json.dumps(
        STRATUM_RULES_V1_DRAFT[
            "transcript"
        ][
            "single_speaker_unresolved"
        ],
        ensure_ascii=False,
        indent=2,
    )
)


print(
    "\nRepresentative repaired candidates:"
)

for item in (
    candidate_pools[
        "transcript"
    ][
        "single_speaker_unresolved"
    ][:10]
):

    print(
        json.dumps(
            item,
            ensure_ascii=False,
            sort_keys=True,
        )
    )


print(
    "\n=== TRUE -> NONE TRANSITION AUDIT ==="
)

print(
    "Total True -> None:",
    sum(
        true_to_none_class_counts.values()
    )
)

for transition_class, count in (
    true_to_none_class_counts.most_common()
):

    print(
        "-",
        transition_class,
        ":",
        count,
    )


print(
    "\n=== TRUE -> NONE BY PROGRAM ==="
)

for program_id in sorted(
    true_to_none_by_program
):

    print(
        program_id,
        ":",
        dict(
            true_to_none_by_program[
                program_id
            ]
        )
    )


print(
    "\n=== REPRESENTATIVE EXPLICIT LABELED CONTEXT ==="
)

for item in (
    true_to_none_class_examples[
        "EXPLICIT_LABELED_CONTEXT"
    ][:10]
):

    print(
        json.dumps(
            item,
            ensure_ascii=False,
            sort_keys=True,
        )
    )


print(
    "\n=== REPRESENTATIVE UNRESOLVED CONTEXT ==="
)

for item in (
    true_to_none_class_examples[
        "UNRESOLVED_CONTEXT"
    ][:10]
):

    print(
        json.dumps(
            item,
            ensure_ascii=False,
            sort_keys=True,
        )
    )


print(
    "\n=== IMPORTANT CURRENT DECISION STATE ==="
)

print(
    "single_speaker_unresolved:",
    candidate_states[
        "transcript"
    ][
        "single_speaker_unresolved"
    ]
)

print(
    "primary_to_nonprimary_transition:",
    candidate_states[
        "transcript"
    ][
        "primary_to_nonprimary_transition"
    ]
)

print(
    "NOTE: True -> None has NOT been relabeled "
    "as confirmed nonprimary."
)


print(
    "\n=== CODE CELL 11 GATES ==="
)

for name, passed in (
    cell_11_gates.items()
):

    print(
        name,
        ":",
        "PASS" if passed else "FAIL"
    )


print(
    "\nCODE CELL 11 GATE:",
    "PASS" if cell_11_pass else "FAIL"
)


assert cell_11_pass, (
    "Code Cell 11 validation failed. "
    "Do not redefine primary_to_nonprimary_transition "
    "until the explicit-context audit is internally consistent."
)
```

<div style="direction: rtl; white-space: normal; line-height: 1.7;">

<h3>Audit Explicit Labeled Context Before Finalizing Primary → Non-primary Transition</h3>

<p>
Audit قبلی نشان داد که از ۶۹۷ Transition با الگوی
<code>primary=True → primary=None</code>،
تعداد ۶۰۴ مورد دارای Label صریح و ۹۳ مورد Unresolved هستند.
</p>

<p>
با این حال، وجود Label به‌تنهایی اثبات نمی‌کند که Utterance سمت راست
واقعاً Non-primary است.
برای مثال Labelهایی مانند «استاد» ممکن است در بعضی Programها نیازمند تفسیر دقیق‌تری باشند.
بنابراین هیچ Labelای صرفاً به دلیل متفاوت بودن متن آن از نام دکتر کمیل رودی
به Non-primary تبدیل نمی‌شود.
</p>

<p>
در این مرحله تمام ۶۰۴ Transition دارای Label صریح بررسی می‌شوند تا مشخص شود:
</p>

<ul>
<li>چه Effective Labelهایی وجود دارند؛</li>
<li>هر Label در کدام Programها دیده می‌شود؛</li>
<li>آیا Effective Canonical Identity برای آن ثبت شده است؛</li>
<li>چه Base Status و Base Rawهایی پشت هر Label قرار دارد؛</li>
<li>آیا همان Label در Recordsی که <code>primary=True</code> هستند نیز دیده شده است یا خیر.</li>
</ul>

<p>
اگر یک Label هم در Primary Records و هم در Null-primary Context دیده شود،
نمی‌توان آن را بدون Evidence بیشتر به‌صورت خودکار Non-primary دانست.
</p>

<p>
این Cell فقط Audit انجام می‌دهد و Rule
<code>primary_to_nonprimary_transition</code>
را هنوز تغییر نمی‌دهد.
</p>

</div>

```python
# Code Cell 12

# ============================================================
# Full explicit-labeled True -> None transition inventory
# ============================================================

explicit_true_to_none = []

for physical_file_id, rows in (
    transcript_files.items()
):

    for left, right in zip(
        rows,
        rows[1:],
    ):

        if not (
            left.get(
                "primary_dr_roudi_corpus"
            )
            is True

            and

            right.get(
                "primary_dr_roudi_corpus"
            )
            is None
        ):
            continue

        right_status = right.get(
            "base_status"
        )

        right_label = right.get(
            "effective_label"
        )

        right_canonical = right.get(
            "effective_canonical"
        )

        right_raw = right.get(
            "base_raw"
        )

        if right_status == "UNRESOLVED":
            continue

        if not (
            right_label is not None
            or right_canonical is not None
            or right_raw is not None
        ):
            continue

        explicit_true_to_none.append(
            {
                "physical_file_id":
                    physical_file_id,

                "program_id":
                    left.get(
                        "program_id"
                    ),

                "left_document_id":
                    left.get(
                        "document_id"
                    ),

                "right_document_id":
                    right.get(
                        "document_id"
                    ),

                "left_utterance_index":
                    left.get(
                        "utterance_index"
                    ),

                "right_utterance_index":
                    right.get(
                        "utterance_index"
                    ),

                "left_effective_label":
                    left.get(
                        "effective_label"
                    ),

                "left_effective_canonical":
                    left.get(
                        "effective_canonical"
                    ),

                "right_base_status":
                    right_status,

                "right_base_raw":
                    right_raw,

                "right_effective_label":
                    right_label,

                "right_effective_canonical":
                    right_canonical,
            }
        )


# ============================================================
# Global transcript label behavior
# ============================================================

label_primary_state_counts = defaultdict(
    Counter
)

label_canonical_counts = defaultdict(
    Counter
)

label_program_counts = defaultdict(
    Counter
)

label_base_status_counts = defaultdict(
    Counter
)

label_base_raw_counts = defaultdict(
    Counter
)


for physical_file_id, rows in (
    transcript_files.items()
):

    for row in rows:

        label = row.get(
            "effective_label"
        )

        if label is None:
            continue

        label_primary_state_counts[
            label
        ][
            scalarize(
                row.get(
                    "primary_dr_roudi_corpus"
                )
            )
        ] += 1

        label_canonical_counts[
            label
        ][
            scalarize(
                row.get(
                    "effective_canonical"
                )
            )
        ] += 1

        label_program_counts[
            label
        ][
            scalarize(
                row.get(
                    "program_id"
                )
            )
        ] += 1

        label_base_status_counts[
            label
        ][
            scalarize(
                row.get(
                    "base_status"
                )
            )
        ] += 1

        label_base_raw_counts[
            label
        ][
            scalarize(
                row.get(
                    "base_raw"
                )
            )
        ] += 1


# ============================================================
# Explicit-transition label inventory
# ============================================================

transition_label_counts = Counter()

transition_label_program_counts = defaultdict(
    Counter
)

transition_label_canonical_counts = defaultdict(
    Counter
)

transition_label_status_counts = defaultdict(
    Counter
)

transition_label_raw_counts = defaultdict(
    Counter
)


for item in explicit_true_to_none:

    label_key = scalarize(
        item[
            "right_effective_label"
        ]
    )

    transition_label_counts[
        label_key
    ] += 1

    transition_label_program_counts[
        label_key
    ][
        scalarize(
            item[
                "program_id"
            ]
        )
    ] += 1

    transition_label_canonical_counts[
        label_key
    ][
        scalarize(
            item[
                "right_effective_canonical"
            ]
        )
    ] += 1

    transition_label_status_counts[
        label_key
    ][
        scalarize(
            item[
                "right_base_status"
            ]
        )
    ] += 1

    transition_label_raw_counts[
        label_key
    ][
        scalarize(
            item[
                "right_base_raw"
            ]
        )
    ] += 1


# ============================================================
# Risk classification
#
# SAFE_EXPLICIT_NONPRIMARY_CANDIDATE:
# label has no observed primary=True records and no canonical Dr.
#
# AMBIGUOUS_LABEL:
# same effective label appears in at least one primary=True record,
# or canonical Dr appears for the label anywhere.
#
# This is an AUDIT classification only.
# ============================================================

DR_CANONICAL_VALUE = (
    "Dr. Komeil Roudi"
)

label_audit = {}

safe_explicit_nonprimary_labels = set()
ambiguous_labels = set()


for label_key in sorted(
    transition_label_counts
):

    # Recover original label value from records.
    matching_transition = next(
        item
        for item
        in explicit_true_to_none
        if scalarize(
            item[
                "right_effective_label"
            ]
        )
        == label_key
    )

    original_label = (
        matching_transition[
            "right_effective_label"
        ]
    )

    global_primary_states = (
        label_primary_state_counts.get(
            original_label,
            Counter(),
        )
    )

    global_canonicals = (
        label_canonical_counts.get(
            original_label,
            Counter(),
        )
    )

    appears_as_primary_true = (
        global_primary_states[
            "True"
        ]
        > 0
    )

    appears_with_dr_canonical = (
        global_canonicals[
            DR_CANONICAL_VALUE
        ]
        > 0
    )

    if (
        not appears_as_primary_true
        and
        not appears_with_dr_canonical
    ):

        audit_status = (
            "SAFE_EXPLICIT_NONPRIMARY_CANDIDATE"
        )

        safe_explicit_nonprimary_labels.add(
            original_label
        )

    else:

        audit_status = (
            "AMBIGUOUS_LABEL"
        )

        ambiguous_labels.add(
            original_label
        )

    label_audit[
        label_key
    ] = {
        "effective_label":
            original_label,

        "transition_count":
            transition_label_counts[
                label_key
            ],

        "transition_programs":
            dict(
                transition_label_program_counts[
                    label_key
                ]
            ),

        "transition_canonicals":
            dict(
                transition_label_canonical_counts[
                    label_key
                ]
            ),

        "transition_base_statuses":
            dict(
                transition_label_status_counts[
                    label_key
                ]
            ),

        "transition_base_raw":
            dict(
                transition_label_raw_counts[
                    label_key
                ]
            ),

        "global_primary_states":
            dict(
                global_primary_states
            ),

        "global_canonicals":
            dict(
                global_canonicals
            ),

        "global_programs":
            dict(
                label_program_counts.get(
                    original_label,
                    Counter(),
                )
            ),

        "appears_as_primary_true":
            appears_as_primary_true,

        "appears_with_dr_canonical":
            appears_with_dr_canonical,

        "audit_status":
            audit_status,
    }


# ============================================================
# Transition-level provisional accounting
# ============================================================

safe_transition_candidates = []
ambiguous_transition_candidates = []


for item in explicit_true_to_none:

    right_label = item.get(
        "right_effective_label"
    )

    if (
        right_label
        in safe_explicit_nonprimary_labels
    ):

        safe_transition_candidates.append(
            {
                **item,
                "audit_status":
                    "SAFE_EXPLICIT_NONPRIMARY_CANDIDATE",
            }
        )

    else:

        ambiguous_transition_candidates.append(
            {
                **item,
                "audit_status":
                    "AMBIGUOUS_LABEL",
            }
        )


# ============================================================
# Validation
# ============================================================

explicit_transition_count_gate = (
    len(
        explicit_true_to_none
    )
    == 604
)

transition_accounting_gate = (
    len(
        safe_transition_candidates
    )
    +
    len(
        ambiguous_transition_candidates
    )
    ==
    len(
        explicit_true_to_none
    )
)

label_accounting_gate = (
    len(
        safe_explicit_nonprimary_labels
    )
    +
    len(
        ambiguous_labels
    )
    ==
    len(
        transition_label_counts
    )
)

no_rule_mutation_gate = (
    candidate_states[
        "transcript"
    ][
        "primary_to_nonprimary_transition"
    ][
        "status"
    ]
    ==
    "EXACT_RULE_ZERO_MATCH_REVIEW_REQUIRED"

    and

    candidate_states[
        "transcript"
    ][
        "primary_to_nonprimary_transition"
    ][
        "candidate_count"
    ]
    == 0
)

no_none_relabel_gate = all(
    item.get(
        "right_effective_label"
    )
    is not None

    for item
    in safe_transition_candidates
)

cell_12_gates = {
    "all_604_explicit_transitions_recovered":
        explicit_transition_count_gate,

    "transition_accounting_complete":
        transition_accounting_gate,

    "label_accounting_complete":
        label_accounting_gate,

    "primary_to_nonprimary_rule_not_mutated":
        no_rule_mutation_gate,

    "safe_candidates_require_explicit_label":
        no_none_relabel_gate,
}


cell_12_pass = all(
    cell_12_gates.values()
)


# ============================================================
# Output
# ============================================================

print(
    "=== EXPLICIT TRUE -> NONE TRANSITIONS ==="
)

print(
    "Recovered explicit labeled transitions:",
    len(
        explicit_true_to_none
    )
)

print(
    "Unique effective labels:",
    len(
        transition_label_counts
    )
)


print(
    "\n=== LABEL AUDIT ==="
)

for label_key in sorted(
    label_audit,
    key=lambda key: (
        -label_audit[
            key
        ][
            "transition_count"
        ],
        key,
    ),
):

    item = label_audit[
        label_key
    ]

    print(
        "\n",
        json.dumps(
            item,
            ensure_ascii=False,
            sort_keys=True,
        )
    )


print(
    "\n=== PROVISIONAL TRANSITION CLASSIFICATION ==="
)

print(
    "Safe explicit non-primary labels:",
    len(
        safe_explicit_nonprimary_labels
    )
)

print(
    "Ambiguous labels:",
    len(
        ambiguous_labels
    )
)

print(
    "Safe explicit transition candidates:",
    len(
        safe_transition_candidates
    )
)

print(
    "Ambiguous transition candidates:",
    len(
        ambiguous_transition_candidates
    )
)


print(
    "\nSafe labels:"
)

for value in sorted(
    scalarize(label)
    for label
    in safe_explicit_nonprimary_labels
):
    print(
        "-",
        value
    )


print(
    "\nAmbiguous labels:"
)

for value in sorted(
    scalarize(label)
    for label
    in ambiguous_labels
):
    print(
        "-",
        value
    )


print(
    "\nRepresentative safe transition candidates:"
)

for item in (
    safe_transition_candidates[:10]
):

    print(
        json.dumps(
            item,
            ensure_ascii=False,
            sort_keys=True,
        )
    )


print(
    "\nRepresentative ambiguous transition candidates:"
)

for item in (
    ambiguous_transition_candidates[:10]
):

    print(
        json.dumps(
            item,
            ensure_ascii=False,
            sort_keys=True,
        )
    )


print(
    "\n=== CODE CELL 12 GATES ==="
)

for name, passed in (
    cell_12_gates.items()
):

    print(
        name,
        ":",
        "PASS"
        if passed
        else "FAIL"
    )


print(
    "\nCODE CELL 12 GATE:",
    "PASS"
    if cell_12_pass
    else "FAIL"
)


assert cell_12_pass, (
    "Explicit labeled transition audit failed. "
    "Do not finalize the primary_to_nonprimary_transition "
    "rule until label identity accounting is complete."
)
```

<div style="direction: rtl; white-space: normal; line-height: 1.7;">

<h3>Finalize Primary → Explicit Non-primary Transition Rule</h3>

<p>
Audit کامل Labelها نشان داد که تمام ۶۰۴ Transition از
<code>primary=True</code> به Context دارای Label صریح، در سه Label
«مجری»، «آقای تهمتن» و «زحمتکش» قرار می‌گیرند.
</p>

<p>
هیچ‌یک از این Labelها در کل Corpus با
<code>primary_dr_roudi_corpus=True</code> مشاهده نشده‌اند و هیچ‌یک
Canonical Identity برابر دکتر کمیل رودی ندارند.
همچنین هیچ Label مبهمی در این Population مشاهده نشد.
</p>

<p>
بنابراین Stratum
<code>primary_to_nonprimary_transition</code>
اکنون می‌تواند با Rule زیر به‌صورت عملیاتی تعریف شود:
</p>

<ul>
<li>دو Utterance باید Adjacent و متعلق به یک Physical File باشند.</li>
<li>Utterance سمت چپ باید صریحاً <code>primary=True</code> باشد.</li>
<li>Utterance سمت راست نباید Unresolved باشد.</li>
<li>سمت راست باید Effective Label صریح داشته باشد.</li>
<li>آن Label در Audit کل Corpus نباید هیچ Record با <code>primary=True</code> داشته باشد.</li>
<li>آن Label نباید در هیچ Record دارای Canonical Identity دکتر کمیل رودی باشد.</li>
</ul>

<p>
مقدار <code>primary=None</code> به‌تنهایی Non-primary تلقی نمی‌شود.
۹۳ Transition از نوع Primary → Unresolved همچنان از این Stratum خارج می‌مانند.
</p>

<p>
این Rule از Evidence فعلی Corpus مشتق شده است و هیچ Speaker Identity جدیدی استنتاج نمی‌کند.
این Cell فقط Rule و Candidate Pool را در حافظه اصلاح می‌کند؛
هیچ Sampling، Question Generation یا Artifact Write انجام نمی‌شود.
</p>

</div>

```python
# Code Cell 13

# ============================================================
# Finalize primary_to_nonprimary_transition from audited
# explicit non-primary identity evidence.
# ============================================================

assert len(
    safe_explicit_nonprimary_labels
) == 3

assert len(
    ambiguous_labels
) == 0

assert len(
    safe_transition_candidates
) == 604


# ============================================================
# Freeze the operational rule in memory
# ============================================================

STRATUM_RULES_V1_DRAFT[
    "transcript"
][
    "primary_to_nonprimary_transition"
] = {
    "eligibility":
        "AUTO_ELIGIBLE_WITH_AUDITED_EXPLICIT_NONPRIMARY_IDENTITY",

    "unit":
        "ADJACENT_UTTERANCE_PAIR",

    "conditions": {
        "same_physical_file":
            True,

        "left_primary_dr_roudi_corpus":
            True,

        "right_primary_dr_roudi_corpus":
            None,

        "right_base_status_not":
            "UNRESOLVED",

        "right_effective_label_required":
            True,

        "right_label_never_observed_primary_true":
            True,

        "right_label_never_observed_with_dr_canonical":
            True,
    },

    "audited_safe_labels":
        sorted(
            safe_explicit_nonprimary_labels
        ),

    "excluded_context": {
        "primary_to_unresolved":
            "EXCLUDED",

        "null_primary_without_explicit_identity":
            "EXCLUDED",

        "ambiguous_label":
            "EXCLUDED_OR_REVIEW_REQUIRED",
    },

    "unknown_speaker_inference":
        "FORBIDDEN",

    "none_means_nonprimary":
        False,

    "evidence_basis": [
        "CODE_CELL_10_PRIMARY_STATE_AUDIT",
        "CODE_CELL_11_TRUE_TO_NONE_CONTEXT_CLASSIFICATION",
        "CODE_CELL_12_GLOBAL_LABEL_IDENTITY_AUDIT",
    ],
}


# ============================================================
# Rebuild exact candidate pool under finalized rule
# ============================================================

candidate_pools[
    "transcript"
][
    "primary_to_nonprimary_transition"
] = []


for physical_file_id, rows in (
    transcript_files.items()
):

    for left, right in zip(
        rows,
        rows[1:],
    ):

        if (
            left.get(
                "primary_dr_roudi_corpus"
            )
            is not True
        ):
            continue

        if (
            right.get(
                "primary_dr_roudi_corpus"
            )
            is not None
        ):
            continue

        if (
            right.get(
                "base_status"
            )
            == "UNRESOLVED"
        ):
            continue

        right_label = right.get(
            "effective_label"
        )

        if (
            right_label
            not in
            safe_explicit_nonprimary_labels
        ):
            continue

        global_primary_states = (
            label_primary_state_counts[
                right_label
            ]
        )

        global_canonicals = (
            label_canonical_counts[
                right_label
            ]
        )

        if (
            global_primary_states[
                "True"
            ]
            > 0
        ):
            continue

        if (
            global_canonicals[
                "Dr. Komeil Roudi"
            ]
            > 0
        ):
            continue

        candidate_pools[
            "transcript"
        ][
            "primary_to_nonprimary_transition"
        ].append(
            {
                "physical_file_id":
                    physical_file_id,

                "program_id":
                    left.get(
                        "program_id"
                    ),

                "left_document_id":
                    left.get(
                        "document_id"
                    ),

                "right_document_id":
                    right.get(
                        "document_id"
                    ),

                "left_utterance_index":
                    left.get(
                        "utterance_index"
                    ),

                "right_utterance_index":
                    right.get(
                        "utterance_index"
                    ),

                "left_primary":
                    True,

                "left_effective_label":
                    left.get(
                        "effective_label"
                    ),

                "left_effective_canonical":
                    left.get(
                        "effective_canonical"
                    ),

                "right_primary":
                    None,

                "right_base_status":
                    right.get(
                        "base_status"
                    ),

                "right_base_raw":
                    right.get(
                        "base_raw"
                    ),

                "right_effective_label":
                    right_label,

                "right_effective_canonical":
                    right.get(
                        "effective_canonical"
                    ),

                "nonprimary_identity_basis":
                    "AUDITED_EXPLICIT_LABEL",
            }
        )


# ============================================================
# Update candidate state
# ============================================================

primary_to_nonprimary_count = len(
    candidate_pools[
        "transcript"
    ][
        "primary_to_nonprimary_transition"
    ]
)


candidate_states[
    "transcript"
][
    "primary_to_nonprimary_transition"
] = {
    "status":
        "EXACT_POOL_READY",

    "candidate_count":
        primary_to_nonprimary_count,

    "rule_status":
        "FINALIZED_FROM_EXPLICIT_NONPRIMARY_LABEL_AUDIT",
}


# ============================================================
# Update rule-state registry in Golden Contract draft
# ============================================================

GOLDEN_DATASET_CONTRACT_V1_DRAFT[
    "stratum_rule_state"
][
    "transcript"
][
    "single_speaker_unresolved"
] = (
    "OPERATIONAL_RULE_VALIDATED"
)

GOLDEN_DATASET_CONTRACT_V1_DRAFT[
    "stratum_rule_state"
][
    "transcript"
][
    "primary_to_nonprimary_transition"
] = (
    "OPERATIONAL_RULE_VALIDATED"
)


# ============================================================
# Independent reconstruction of expected audited population
# ============================================================

audited_transition_identity_set = {
    (
        item[
            "physical_file_id"
        ],
        item[
            "left_document_id"
        ],
        item[
            "right_document_id"
        ],
    )
    for item
    in safe_transition_candidates
}


rebuilt_transition_identity_set = {
    (
        item[
            "physical_file_id"
        ],
        item[
            "left_document_id"
        ],
        item[
            "right_document_id"
        ],
    )
    for item
    in candidate_pools[
        "transcript"
    ][
        "primary_to_nonprimary_transition"
    ]
}


# ============================================================
# All exact-rule population readiness after repairs
# ============================================================

exact_rule_population_status = {}


for source_type, strata in (
    exact_rule_strata.items()
):

    for stratum in sorted(
        strata
    ):

        count = len(
            candidate_pools[
                source_type
            ][
                stratum
            ]
        )

        exact_rule_population_status[
            (
                source_type,
                stratum,
            )
        ] = count


# ============================================================
# Validation
# ============================================================

final_transition_count_gate = (
    primary_to_nonprimary_count
    == 604
)


exact_identity_match_gate = (
    rebuilt_transition_identity_set
    ==
    audited_transition_identity_set
)


no_unresolved_transition_gate = all(
    item[
        "right_base_status"
    ]
    != "UNRESOLVED"

    for item
    in candidate_pools[
        "transcript"
    ][
        "primary_to_nonprimary_transition"
    ]
)


no_dr_identity_on_right_gate = all(
    item[
        "right_effective_label"
    ]
    in safe_explicit_nonprimary_labels

    and

    item[
        "right_effective_canonical"
    ]
    != "Dr. Komeil Roudi"

    for item
    in candidate_pools[
        "transcript"
    ][
        "primary_to_nonprimary_transition"
    ]
)


safe_labels_global_gate = all(
    label_primary_state_counts[
        label
    ][
        "True"
    ]
    == 0

    and

    label_canonical_counts[
        label
    ][
        "Dr. Komeil Roudi"
    ]
    == 0

    for label
    in safe_explicit_nonprimary_labels
)


all_exact_rule_pools_nonempty_gate = all(
    count > 0

    for count
    in exact_rule_population_status.values()
)


single_unresolved_still_seven_gate = (
    len(
        candidate_pools[
            "transcript"
        ][
            "single_speaker_unresolved"
        ]
    )
    == 7
)


rule_state_gate = (
    GOLDEN_DATASET_CONTRACT_V1_DRAFT[
        "stratum_rule_state"
    ][
        "transcript"
    ][
        "single_speaker_unresolved"
    ]
    == "OPERATIONAL_RULE_VALIDATED"

    and

    GOLDEN_DATASET_CONTRACT_V1_DRAFT[
        "stratum_rule_state"
    ][
        "transcript"
    ][
        "primary_to_nonprimary_transition"
    ]
    == "OPERATIONAL_RULE_VALIDATED"
)


no_sampling_gate = (
    GOLDEN_DATASET_CONTRACT_V1_DRAFT[
        "sampling_policy"
    ][
        "status"
    ]
    == "NOT_YET_DEFINED"
)


no_question_generation_gate = (
    GOLDEN_DATASET_CONTRACT_V1_DRAFT[
        "question_answer_policy"
    ][
        "question_generation_status"
    ]
    == "NOT_STARTED"
)


cell_13_gates = {
    "primary_to_nonprimary_count_604":
        final_transition_count_gate,

    "rebuilt_population_matches_audit":
        exact_identity_match_gate,

    "unresolved_context_excluded":
        no_unresolved_transition_gate,

    "dr_identity_excluded_from_right_side":
        no_dr_identity_on_right_gate,

    "audited_safe_labels_globally_nonprimary":
        safe_labels_global_gate,

    "all_exact_rule_candidate_pools_nonempty":
        all_exact_rule_pools_nonempty_gate,

    "single_speaker_unresolved_remains_7":
        single_unresolved_still_seven_gate,

    "repaired_rule_states_validated":
        rule_state_gate,

    "sampling_not_started":
        no_sampling_gate,

    "question_generation_not_started":
        no_question_generation_gate,
}


cell_13_pass = all(
    cell_13_gates.values()
)


# ============================================================
# Output
# ============================================================

print(
    "=== FINALIZED RULE: primary_to_nonprimary_transition ==="
)

print(
    json.dumps(
        STRATUM_RULES_V1_DRAFT[
            "transcript"
        ][
            "primary_to_nonprimary_transition"
        ],
        ensure_ascii=False,
        indent=2,
    )
)


print(
    "\nCandidate count:",
    primary_to_nonprimary_count
)


print(
    "\nCandidate distribution by right label:"
)

right_label_counts = Counter(
    item[
        "right_effective_label"
    ]
    for item
    in candidate_pools[
        "transcript"
    ][
        "primary_to_nonprimary_transition"
    ]
)

for label, count in (
    right_label_counts.most_common()
):
    print(
        "-",
        label,
        ":",
        count,
    )


print(
    "\nCandidate distribution by program:"
)

transition_program_counts_final = Counter(
    item[
        "program_id"
    ]
    for item
    in candidate_pools[
        "transcript"
    ][
        "primary_to_nonprimary_transition"
    ]
)

for program_id, count in (
    transition_program_counts_final.most_common()
):
    print(
        "-",
        program_id,
        ":",
        count,
    )


print(
    "\nRepresentative finalized candidates:"
)

for item in (
    candidate_pools[
        "transcript"
    ][
        "primary_to_nonprimary_transition"
    ][:10]
):
    print(
        json.dumps(
            item,
            ensure_ascii=False,
            sort_keys=True,
        )
    )


print(
    "\n=== ALL EXACT-RULE POOL COUNTS ==="
)

for (
    source_type,
    stratum
), count in sorted(
    exact_rule_population_status.items()
):

    print(
        "-",
        source_type,
        "/",
        stratum,
        ":",
        count,
    )


print(
    "\n=== CODE CELL 13 GATES ==="
)

for name, passed in (
    cell_13_gates.items()
):

    print(
        name,
        ":",
        "PASS" if passed else "FAIL"
    )


print(
    "\nCODE CELL 13 GATE:",
    "PASS" if cell_13_pass else "FAIL"
)


assert cell_13_pass, (
    "Final primary_to_nonprimary_transition rule validation failed. "
    "Do not proceed to dependency-stratum measurement until "
    "the audited transition population is reproduced exactly."
)
```

```python
# Code Cell 14

# ============================================================
# Dependency Resolution Readiness
#
# REPAIR v2:
# - Do not assume ANY required_paths key for NB09.
# - Discover NB09 / dedup evidence from actual registered paths,
#   repository files, and NB09 notebook if necessary.
# - Do not assume all NB11 implementation terms live
#   in the same code cell.
#
# READ-ONLY ONLY
# No candidate generation.
# No artifact writes.
# No network access.
# ============================================================


DEPENDENCY_STRATA = {
    "website": [
        "citation_locator_sensitive_case",
    ],

    "book": [
        "intrinsically_tiny_narrative_region",
        "intrinsically_tiny_table_region",
        "long_section",
        "large_table_run",
    ],

    "transcript": [
        "long_utterance_over_500",
        "long_utterance_over_800",
        "token_fallback_sentence_case",
    ],
}


# ============================================================
# Local helpers
# ============================================================

def safe_path(value):

    if isinstance(value, Path):
        return value

    if isinstance(value, str):
        return Path(value)

    return None


def recursive_signal_matches_local(
    obj,
    terms,
    prefix="",
    output=None,
    max_value_chars=240,
):

    if output is None:
        output = []

    lowered_terms = {
        str(term).lower()
        for term in terms
    }

    if isinstance(obj, dict):

        for key, value in obj.items():

            path = (
                f"{prefix}.{key}"
                if prefix
                else str(key)
            )

            key_lower = str(
                key
            ).lower()

            if any(
                term in key_lower
                for term
                in lowered_terms
            ):

                rendered = repr(
                    value
                )

                if (
                    len(rendered)
                    > max_value_chars
                ):
                    rendered = (
                        rendered[
                            :max_value_chars
                        ]
                        + "...<TRUNCATED>"
                    )

                output.append(
                    (
                        path,
                        rendered,
                    )
                )

            recursive_signal_matches_local(
                value,
                terms,
                prefix=path,
                output=output,
                max_value_chars=(
                    max_value_chars
                ),
            )

    elif isinstance(obj, list):

        for index, value in enumerate(
            obj
        ):

            path = (
                f"{prefix}[{index}]"
            )

            recursive_signal_matches_local(
                value,
                terms,
                prefix=path,
                output=output,
                max_value_chars=(
                    max_value_chars
                ),
            )

    elif obj is not None:

        rendered = str(
            obj
        )

        rendered_lower = (
            rendered.lower()
        )

        if any(
            term in rendered_lower
            for term
            in lowered_terms
        ):

            if (
                len(rendered)
                > max_value_chars
            ):
                rendered = (
                    rendered[
                        :max_value_chars
                    ]
                    + "...<TRUNCATED>"
                )

            output.append(
                (
                    prefix,
                    rendered,
                )
            )

    return output


def inspect_json_or_jsonl(
    path,
    max_jsonl_rows=100,
):

    result = {
        "path":
            path,

        "format":
            path.suffix.lower(),

        "parsed":
            False,

        "payload":
            None,

        "error":
            None,
    }

    try:

        if (
            path.suffix.lower()
            == ".json"
        ):

            with path.open(
                "r",
                encoding="utf-8",
            ) as f:

                payload = json.load(
                    f
                )

            result[
                "payload"
            ] = payload

            result[
                "parsed"
            ] = True

        elif (
            path.suffix.lower()
            == ".jsonl"
        ):

            rows = []

            with path.open(
                "r",
                encoding="utf-8",
            ) as f:

                for line_number, line in enumerate(
                    f,
                    start=1,
                ):

                    if (
                        line_number
                        > max_jsonl_rows
                    ):
                        break

                    line = line.strip()

                    if not line:
                        continue

                    rows.append(
                        json.loads(
                            line
                        )
                    )

            result[
                "payload"
            ] = rows

            result[
                "parsed"
            ] = True

    except Exception as exc:

        result[
            "error"
        ] = {
            "type":
                type(
                    exc
                ).__name__,

            "message":
                str(
                    exc
                )[:500],
        }

    return result


# ============================================================
# 1) Inspect ACTUAL required_paths registry
# ============================================================

print(
    "=== REQUIRED_PATHS REGISTRY INSPECTION ==="
)

print(
    "Registered keys:",
    len(
        required_paths
    )
)


for key in sorted(
    required_paths
):

    value = required_paths[
        key
    ]

    print(
        "-",
        key,
        "->",
        value,
    )


# ============================================================
# 2) Discover possible NB09 / dedup evidence from
#    registered paths without assuming key names.
# ============================================================

nb09_path_terms = {
    "nb09",
    "09_",
    "dedup",
    "duplicate",
    "quality",
    "similar",
    "near",
    "cross_source",
    "cross-source",
}


registered_nb09_candidates = []


for key, value in (
    required_paths.items()
):

    path = safe_path(
        value
    )

    if path is None:
        continue

    searchable = (
        str(key)
        + " "
        + str(path)
    ).lower()

    if any(
        term in searchable
        for term
        in nb09_path_terms
    ):

        registered_nb09_candidates.append(
            {
                "key":
                    key,

                "path":
                    path,

                "exists":
                    path.exists(),

                "is_file":
                    path.is_file(),
            }
        )


# ============================================================
# 3) Repository-level NB09 / duplicate evidence discovery
# ============================================================

search_roots = [
    repo_root
    / "artifacts"
    / "phase2",

    repo_root
    / "reports"
    / "phase2",

    repo_root
    / "data"
    / "phase2",

    repo_root
    / "notebooks"
    / "phase2",
]


discovered_nb09_files = []


allowed_suffixes = {
    ".json",
    ".jsonl",
    ".ipynb",
}


for root in search_roots:

    if not root.exists():
        continue

    for path in root.rglob(
        "*"
    ):

        if not path.is_file():
            continue

        if (
            ".ipynb_checkpoints"
            in path.parts
        ):
            continue

        if (
            path.suffix.lower()
            not in allowed_suffixes
        ):
            continue

        relative_text = str(
            path.relative_to(
                repo_root
            )
        ).lower()

        if any(
            term in relative_text
            for term
            in nb09_path_terms
        ):

            discovered_nb09_files.append(
                path
            )


discovered_nb09_files = sorted(
    set(
        discovered_nb09_files
    )
)


# ============================================================
# 4) Find the actual NB09 notebook separately
# ============================================================

nb09_notebook_candidates = []


phase2_notebook_dir = (
    repo_root
    / "notebooks"
    / "phase2"
)


if phase2_notebook_dir.is_dir():

    for path in sorted(
        phase2_notebook_dir.glob(
            "09*.ipynb"
        )
    ):

        if path.is_file():

            nb09_notebook_candidates.append(
                path
            )


# ============================================================
# 5) Parse candidate JSON / JSONL evidence files
# ============================================================

json_evidence_candidates = [
    path
    for path
    in discovered_nb09_files
    if (
        path.suffix.lower()
        in {
            ".json",
            ".jsonl",
        }
    )
]


parsed_nb09_evidence = []


duplicate_signal_terms = {
    "duplicate",
    "dedup",
    "near_duplicate",
    "near-duplicate",
    "similarity",
    "similar",
    "cross_source",
    "cross-source",
    "group",
    "candidate",
}


for path in (
    json_evidence_candidates
):

    inspection = (
        inspect_json_or_jsonl(
            path
        )
    )

    signal_matches = []

    if inspection[
        "parsed"
    ]:

        signal_matches = (
            recursive_signal_matches_local(
                inspection[
                    "payload"
                ],
                duplicate_signal_terms,
            )
        )

    parsed_nb09_evidence.append(
        {
            "path":
                path,

            "parsed":
                inspection[
                    "parsed"
                ],

            "error":
                inspection[
                    "error"
                ],

            "signal_match_count":
                len(
                    signal_matches
                ),

            "signal_matches":
                signal_matches[
                    :30
                ],
        }
    )


# ============================================================
# 6) Inspect NB09 notebook source/output for duplicate evidence
# ============================================================

nb09_notebook_signal_matches = []


for path in (
    nb09_notebook_candidates
):

    try:

        with path.open(
            "r",
            encoding="utf-8",
        ) as f:

            notebook_obj = (
                json.load(
                    f
                )
            )

        notebook_matches = (
            recursive_signal_matches_local(
                notebook_obj,
                duplicate_signal_terms,
            )
        )

        nb09_notebook_signal_matches.append(
            {
                "path":
                    path,

                "match_count":
                    len(
                        notebook_matches
                    ),

                "matches":
                    notebook_matches[
                        :50
                    ],
            }
        )

    except Exception as exc:

        nb09_notebook_signal_matches.append(
            {
                "path":
                    path,

                "match_count":
                    0,

                "matches":
                    [],

                "error":
                    {
                        "type":
                            type(
                                exc
                            ).__name__,

                        "message":
                            str(
                                exc
                            )[:500],
                    },
            }
        )


# ============================================================
# 7) Recover NB11 implementation evidence
# ============================================================

nb11_code_sources = [
    cell_source(
        cell
    )
    for cell
    in code_cells
]


implementation_needles = {

    "book_section_measurement": [
        "book_section_token_counts",
        "book_section_stats",
        "book_sections_over_800",
    ],

    "book_table_run_measurement": [
        "book_table_runs",
        "table_run_token_counts",
    ],

    "book_tiny_root_cause": [
        "BOOK_TINY_ROOT_CAUSE",
        "whole_region_under_100",
    ],

    "transcript_long_utterance": [
        "transcript_long_utterances",
        "long_utterances_with_sentence_over_500",
    ],

    "transcript_fallback": [
        "token_offset_windows",
        "transcript_token_fallback_fragment",
    ],

    "token_measurement_function": [
        "def measurement_token_ids",
        "def token_count_text",
    ],
}


implementation_evidence = {}


for evidence_name, needles in (
    implementation_needles.items()
):

    needle_locations = {}

    for needle in needles:

        positions = []

        for code_position, source in enumerate(
            nb11_code_sources,
            start=1,
        ):

            if (
                needle.lower()
                in source.lower()
            ):

                positions.append(
                    code_position
                )

        needle_locations[
            needle
        ] = positions

    implementation_evidence[
        evidence_name
    ] = needle_locations


implementation_group_complete = {
    evidence_name:
        all(
            bool(
                positions
            )
            for positions
            in needle_map.values()
        )

    for evidence_name, needle_map
    in implementation_evidence.items()
}


# ============================================================
# 8) Frozen measurement-tokenizer identity
# ============================================================

measurement_tokenizer_record = (
    parsed_json[
        "chunk_contract"
    ].get(
        "measurement_tokenizer"
    )
    or {}
)


measurement_model_name = (
    measurement_tokenizer_record.get(
        "model_name"
    )
)


expected_measurement_model = (
    "intfloat/multilingual-e5-small"
)


# ============================================================
# 9) Discover local tokenizer snapshots
# ============================================================

hf_model_cache = (
    Path.home()
    / ".cache"
    / "huggingface"
    / "hub"
    / "models--intfloat--multilingual-e5-small"
    / "snapshots"
)


local_e5_snapshots = []


if hf_model_cache.is_dir():

    for snapshot_dir in sorted(
        hf_model_cache.iterdir()
    ):

        if not snapshot_dir.is_dir():
            continue

        tokenizer_assets = sorted(
            path.name
            for path
            in snapshot_dir.iterdir()
            if (
                path.is_file()

                and

                path.name
                in {
                    "tokenizer.json",
                    "tokenizer_config.json",
                    "special_tokens_map.json",
                    "sentencepiece.bpe.model",
                    "vocab.json",
                    "vocab.txt",
                    "merges.txt",
                }
            )
        )

        local_e5_snapshots.append(
            {
                "snapshot":
                    snapshot_dir.name,

                "path":
                    snapshot_dir,

                "tokenizer_assets":
                    tokenizer_assets,

                "has_tokenizer_json":
                    (
                        "tokenizer.json"
                        in tokenizer_assets
                    ),

                "has_config":
                    (
                        "tokenizer_config.json"
                        in tokenizer_assets
                    ),
            }
        )


usable_local_snapshots = [
    item
    for item
    in local_e5_snapshots
    if (
        item[
            "has_tokenizer_json"
        ]

        and

        item[
            "has_config"
        ]
    )
]


# ============================================================
# 10) Recover snapshot IDs declared in NB11
# ============================================================

snapshot_pattern = re.compile(
    r'["\']snapshot["\']\s*:\s*'
    r'["\']([0-9a-f]{40,64})["\']'
)


nb11_declared_snapshots = set()


for source in (
    nb11_code_sources
):

    for match in (
        snapshot_pattern.finditer(
            source
        )
    ):

        nb11_declared_snapshots.add(
            match.group(1)
        )


local_snapshot_ids = {
    item[
        "snapshot"
    ]
    for item
    in usable_local_snapshots
}


declared_local_overlap = sorted(
    nb11_declared_snapshots
    &
    local_snapshot_ids
)


# ============================================================
# 11) Dependency-state invariants
# ============================================================

dependency_state_gate = all(
    candidate_states[
        source_type
    ][
        stratum
    ][
        "status"
    ]
    == "DEPENDENCY_REQUIRED"

    and

    candidate_states[
        source_type
    ][
        stratum
    ][
        "candidate_count"
    ]
    is None

    for source_type, strata
    in DEPENDENCY_STRATA.items()

    for stratum
    in strata
)


# ============================================================
# 12) Evidence-readiness calculations
# ============================================================

registered_nb09_existing = [
    item
    for item
    in registered_nb09_candidates
    if item[
        "exists"
    ]
]


parsed_duplicate_signal_files = [
    item
    for item
    in parsed_nb09_evidence
    if (
        item[
            "parsed"
        ]

        and

        item[
            "signal_match_count"
        ]
        > 0
    )
]


nb09_notebook_with_signals = [
    item
    for item
    in nb09_notebook_signal_matches
    if (
        item[
            "match_count"
        ]
        > 0
    )
]


nb09_evidence_discoverable = (
    len(
        parsed_duplicate_signal_files
    )
    > 0

    or

    len(
        nb09_notebook_with_signals
    )
    > 0
)


# ============================================================
# 13) Readiness gates
# ============================================================

readiness_gates = {

    "all_8_dependencies_still_explicit":
        dependency_state_gate,

    "required_paths_registry_inspected":
        isinstance(
            required_paths,
            dict,
        ),

    "nb09_notebook_discovered":
        len(
            nb09_notebook_candidates
        )
        > 0,

    "nb09_duplicate_similarity_evidence_discoverable":
        nb09_evidence_discoverable,

    "nb11_book_section_measurement_logic_found":
        implementation_group_complete[
            "book_section_measurement"
        ],

    "nb11_table_run_measurement_logic_found":
        implementation_group_complete[
            "book_table_run_measurement"
        ],

    "nb11_tiny_region_root_cause_logic_found":
        implementation_group_complete[
            "book_tiny_root_cause"
        ],

    "nb11_long_utterance_logic_found":
        implementation_group_complete[
            "transcript_long_utterance"
        ],

    "nb11_fallback_logic_found":
        implementation_group_complete[
            "transcript_fallback"
        ],

    "nb11_token_measurement_functions_found":
        implementation_group_complete[
            "token_measurement_function"
        ],

    "measurement_tokenizer_identity":
        (
            measurement_model_name
            ==
            expected_measurement_model
        ),

    "local_tokenizer_assets_available":
        len(
            usable_local_snapshots
        )
        > 0,

    "nb11_declared_snapshot_available_locally":
        len(
            declared_local_overlap
        )
        > 0,

    "dependency_candidate_counts_not_fabricated":
        all(
            candidate_states[
                source_type
            ][
                stratum
            ][
                "candidate_count"
            ]
            is None

            for source_type, strata
            in DEPENDENCY_STRATA.items()

            for stratum
            in strata
        ),

    "sampling_not_started":
        (
            GOLDEN_DATASET_CONTRACT_V1_DRAFT[
                "sampling_policy"
            ][
                "status"
            ]
            == "NOT_YET_DEFINED"
        ),

    "question_generation_not_started":
        (
            GOLDEN_DATASET_CONTRACT_V1_DRAFT[
                "question_answer_policy"
            ][
                "question_generation_status"
            ]
            == "NOT_STARTED"
        ),
}


readiness_pass = all(
    readiness_gates.values()
)


# ============================================================
# Output
# ============================================================

print(
    "\n=== REGISTERED NB09-LIKE PATHS ==="
)

if registered_nb09_candidates:

    for item in (
        registered_nb09_candidates
    ):

        print(
            "- key:",
            item[
                "key"
            ],
            "| path:",
            item[
                "path"
            ],
            "| exists:",
            item[
                "exists"
            ],
            "| file:",
            item[
                "is_file"
            ],
        )

else:

    print(
        "NONE_REGISTERED"
    )


print(
    "\n=== DISCOVERED NB09 / DEDUP FILES ==="
)

print(
    "Files:",
    len(
        discovered_nb09_files
    )
)


for path in (
    discovered_nb09_files[:60]
):

    print(
        "-",
        path.relative_to(
            repo_root
        )
    )


print(
    "\n=== NB09 NOTEBOOK CANDIDATES ==="
)

for path in (
    nb09_notebook_candidates
):

    print(
        "-",
        path.relative_to(
            repo_root
        )
    )


print(
    "\n=== PARSED NB09 JSON / JSONL EVIDENCE ==="
)

for item in (
    parsed_nb09_evidence
):

    print(
        "-",
        item[
            "path"
        ].relative_to(
            repo_root
        ),
        "| parsed:",
        item[
            "parsed"
        ],
        "| duplicate/similarity signal matches:",
        item[
            "signal_match_count"
        ],
    )

    if item[
        "error"
    ]:

        print(
            "  error:",
            item[
                "error"
            ]
        )


print(
    "\n=== NB09 NOTEBOOK SIGNAL SUMMARY ==="
)

for item in (
    nb09_notebook_signal_matches
):

    print(
        "-",
        item[
            "path"
        ].relative_to(
            repo_root
        ),
        "| signal matches:",
        item[
            "match_count"
        ],
    )

    for field_path, value in (
        item[
            "matches"
        ][:20]
    ):

        print(
            "   ",
            field_path,
            "=",
            value,
        )


print(
    "\n=== NB11 IMPLEMENTATION EVIDENCE ==="
)

for evidence_name, needle_map in (
    implementation_evidence.items()
):

    print(
        f"\n[{evidence_name}]"
    )

    for needle, positions in (
        needle_map.items()
    ):

        print(
            "-",
            needle,
            "->",
            positions,
        )

    print(
        "complete:",
        implementation_group_complete[
            evidence_name
        ],
    )


print(
    "\n=== MEASUREMENT TOKENIZER ==="
)

print(
    "Expected model:",
    expected_measurement_model
)

print(
    "Chunk-contract model:",
    measurement_model_name
)

print(
    "Local cache root:",
    hf_model_cache
)

print(
    "Usable local snapshots:",
    len(
        usable_local_snapshots
    )
)


for item in (
    usable_local_snapshots
):

    print(
        "-",
        item[
            "snapshot"
        ],
        "| assets:",
        item[
            "tokenizer_assets"
        ],
    )


print(
    "NB11 declared snapshots:",
    sorted(
        nb11_declared_snapshots
    )
)

print(
    "Declared/local overlap:",
    declared_local_overlap
)


print(
    "\n=== DEPENDENCY READINESS GATES ==="
)


for name, passed in (
    readiness_gates.items()
):

    print(
        name,
        ":",
        "PASS"
        if passed
        else "FAIL"
    )


print(
    "\nCODE CELL 14 GATE:",
    "PASS"
    if readiness_pass
    else "FAIL"
)


assert readiness_pass, (
    "Dependency resolution readiness failed. "
    "Do not generate dependency candidate pools until "
    "all required evidence and frozen tokenizer assets "
    "are verified from actual repository state."
)
```

<p>
Dependency Readiness با موفقیت PASS شد.
اکنون قبل از اجرای Measurement واقعی، Schema و Identity Linkage منابع وابسته بررسی می‌شود
تا هیچ Candidate بر اساس Field یا Mapping فرضی ساخته نشود.
</p>

برای Website باید مشخص شود Exact/Near/Cross-source Relation Artifactهای NB09
دقیقاً چه شناسه‌هایی برای طرفین Relation نگهداری می‌کنند و کدام Evidence
برای ساخت یک Citation-sensitive Case قابل بازیابی است.
</p>

<p>
برای Book و Transcript نیز Code و Output واقعی Cellهای Measurement در NB11 بررسی می‌شوند
تا مشخص شود آیا Objectهای محاسبه‌شده شامل
<code>document_id</code>، <code>section identity</code>،
<code>table run identity</code>، <code>utterance identity</code>
یا <code>fallback linkage</code> هستند یا باید در NB12 دوباره Measurement شوند.

<p>
این مرحله Read-only است.
Tokenizer Load نمی‌شود، هیچ Candidate Pool وابسته ساخته نمی‌شود،
هیچ Count جدیدی Freeze نمی‌شود و هیچ Artifactی نوشته نمی‌شود.
</p>

<h4>معیار پایان</h4>

<ul>
<li>Schema واقعی سه Artifact رابطه‌ای NB09 مشخص شود.</li>
<li>Fieldهای هویت دو طرف Relation بازیابی شوند.</li>
<li>NB11 Measurement Objectها و Fieldهای هویتی آن‌ها مشخص شوند.</li>
<li>برای هر یک از هشت Dependency تعیین شود: <code>REUSE_PRECOMPUTED_EVIDENCE</code> یا <code>REMEASURE_IN_NB12</code>.</li>
</ul>

</div>

```python
# Code Cell 15

# ============================================================
# Dependency Evidence Schema & Identity Linkage Inspection
#
# Read-only.
# No tokenizer loading.
# No candidate generation.
# No artifact writes.
# ============================================================


# ============================================================
# 1) Resolve exact NB09 relation artifacts from files that
#    were ACTUALLY discovered in Code Cell 14.
# ============================================================

relation_file_names = {
    "exact":
        "dedup_exact_relations_v1.jsonl",

    "near":
        "dedup_near_candidates_v1.jsonl",

    "cross_source":
        "dedup_cross_source_candidates_v1.jsonl",
}


resolved_relation_files = {}


for relation_type, filename in (
    relation_file_names.items()
):

    matches = [
        path
        for path
        in discovered_nb09_files
        if (
            path.name
            == filename
        )
    ]

    if len(matches) == 1:

        resolved_relation_files[
            relation_type
        ] = matches[0]

    else:

        resolved_relation_files[
            relation_type
        ] = None


# ============================================================
# 2) Inspect actual schemas and a few identity-only examples
# ============================================================

relation_schemas = {}

relation_identity_fields = {}

relation_examples = {}


identity_terms = {
    "document",
    "source",
    "left",
    "right",
    "a_id",
    "b_id",
    "id_a",
    "id_b",
    "member",
    "group",
    "pair",
    "hash",
    "similarity",
    "score",
    "relation",
    "type",
}


for relation_type, path in (
    resolved_relation_files.items()
):

    relation_schemas[
        relation_type
    ] = set()

    relation_identity_fields[
        relation_type
    ] = Counter()

    relation_examples[
        relation_type
    ] = []

    if path is None:
        continue

    with path.open(
        "r",
        encoding="utf-8",
    ) as f:

        for row_number, line in enumerate(
            f,
            start=1,
        ):

            line = line.strip()

            if not line:
                continue

            row = json.loads(
                line
            )

            relation_schemas[
                relation_type
            ].update(
                row.keys()
            )

            flat_fields = (
                flatten_non_text_fields(
                    row
                )
            )

            identity_only = {}

            for field_path, value in (
                flat_fields
            ):

                lowered = (
                    field_path.lower()
                )

                if any(
                    term in lowered
                    for term
                    in identity_terms
                ):

                    relation_identity_fields[
                        relation_type
                    ][
                        field_path
                    ] += 1

                    identity_only[
                        field_path
                    ] = value

            if (
                len(
                    relation_examples[
                        relation_type
                    ]
                )
                < 3
            ):

                relation_examples[
                    relation_type
                ].append(
                    identity_only
                )

            if row_number >= 500:
                break


# ============================================================
# 3) Inspect NB11 relevant implementation cells
# ============================================================

relevant_nb11_positions = sorted(
    {
        position
        for evidence_map
        in implementation_evidence.values()

        for positions
        in evidence_map.values()

        for position
        in positions
    }
)


nb11_relevant_code = {}


for position in (
    relevant_nb11_positions
):

    source = nb11_code_sources[
        position - 1
    ]

    nb11_relevant_code[
        position
    ] = source


# ============================================================
# 4) Search exact identity-bearing structures in NB11 code
# ============================================================

nb11_identity_needles = {
    "book_section_identity": [
        "section_key",
        "heading_path",
        "work_id",
        "edition_id",
        "document_id",
    ],

    "book_table_run_identity": [
        "table_run",
        "document_id",
        "paragraph",
        "work_id",
        "edition_id",
    ],

    "book_tiny_region_identity": [
        "whole_region_under_100",
        "document_id",
        "source_document",
        "book",
    ],

    "transcript_long_identity": [
        "transcript_long_utterances",
        "document_id",
        "utterance",
        "token_count",
    ],

    "transcript_fallback_identity": [
        "transcript_token_fallback_fragment",
        "source_document",
        "document_id",
        "split_method",
    ],
}


nb11_identity_hits = {}


for group_name, needles in (
    nb11_identity_needles.items()
):

    needle_hits = {}

    for needle in needles:

        positions = []

        for position, source in (
            nb11_relevant_code.items()
        ):

            if (
                needle.lower()
                in source.lower()
            ):

                positions.append(
                    position
                )

        needle_hits[
            needle
        ] = positions

    nb11_identity_hits[
        group_name
    ] = needle_hits


# ============================================================
# 5) Inspect outputs of relevant NB11 code cells
# ============================================================

nb11_relevant_outputs = {}


for position in (
    relevant_nb11_positions
):

    cell = code_cells[
        position - 1
    ]

    output_text_parts = []

    for output in cell.get(
        "outputs",
        []
    ):

        output_type = output.get(
            "output_type"
        )

        if output_type == "stream":

            text_value = output.get(
                "text",
                []
            )

            if isinstance(
                text_value,
                list,
            ):
                output_text_parts.append(
                    "".join(
                        text_value
                    )
                )

            else:
                output_text_parts.append(
                    str(
                        text_value
                    )
                )

        elif output_type in {
            "execute_result",
            "display_data",
        }:

            data = output.get(
                "data",
                {}
            )

            text_plain = data.get(
                "text/plain"
            )

            if text_plain is not None:

                if isinstance(
                    text_plain,
                    list,
                ):
                    output_text_parts.append(
                        "".join(
                            text_plain
                        )
                    )

                else:
                    output_text_parts.append(
                        str(
                            text_plain
                        )
                    )

    combined_output = "\n".join(
        output_text_parts
    )

    nb11_relevant_outputs[
        position
    ] = combined_output


# ============================================================
# 6) Find dependency-specific output signals
# ============================================================

dependency_output_terms = {
    "book_section": [
        "section",
        "over_800",
        "token",
    ],

    "book_table_run": [
        "table",
        "run",
        "token",
    ],

    "book_tiny": [
        "under_100",
        "tiny",
        "region",
    ],

    "transcript_long": [
        "long",
        "utterance",
        "500",
    ],

    "transcript_fallback": [
        "fallback",
        "sentence",
        "token",
    ],
}


dependency_output_hits = {}


for dependency_name, terms in (
    dependency_output_terms.items()
):

    matches = []

    for position, output_text in (
        nb11_relevant_outputs.items()
    ):

        lowered = (
            output_text.lower()
        )

        if all(
            term.lower()
            in lowered
            for term
            in terms
        ):

            matches.append(
                position
            )

    dependency_output_hits[
        dependency_name
    ] = matches


# ============================================================
# 7) Determine reuse vs remeasurement conservatively
# ============================================================

dependency_resolution_plan = {
    "website": {
        "citation_locator_sensitive_case":
            None,
    },

    "book": {
        "intrinsically_tiny_narrative_region":
            None,

        "intrinsically_tiny_table_region":
            None,

        "long_section":
            None,

        "large_table_run":
            None,
    },

    "transcript": {
        "long_utterance_over_500":
            None,

        "long_utterance_over_800":
            None,

        "token_fallback_sentence_case":
            None,
    },
}


# ------------------------------------------------------------
# Website:
# We can reuse only if actual relation artifacts expose
# recoverable two-sided identities.
# ------------------------------------------------------------

relation_identity_available = all(
    resolved_relation_files[
        relation_type
    ]
    is not None

    and

    len(
        relation_identity_fields[
            relation_type
        ]
    )
    > 0

    for relation_type
    in [
        "exact",
        "near",
        "cross_source",
    ]
)


dependency_resolution_plan[
    "website"
][
    "citation_locator_sensitive_case"
] = (
    "REUSE_PRECOMPUTED_EVIDENCE"
    if relation_identity_available
    else
    "INSUFFICIENT_IDENTITY_EVIDENCE"
)


# ------------------------------------------------------------
# Book / Transcript:
# Do NOT assume precomputed objects are reusable merely because
# variable names exist. Reuse requires identity-bearing evidence.
# Otherwise NB12 must remeasure deterministically.
# ------------------------------------------------------------

book_section_identity_ready = all(
    bool(
        nb11_identity_hits[
            "book_section_identity"
        ][needle]
    )
    for needle
    in [
        "section_key",
        "heading_path",
        "work_id",
        "edition_id",
        "document_id",
    ]
)


book_table_identity_ready = (
    bool(
        nb11_identity_hits[
            "book_table_run_identity"
        ][
            "table_run"
        ]
    )

    and

    bool(
        nb11_identity_hits[
            "book_table_run_identity"
        ][
            "document_id"
        ]
    )
)


book_tiny_identity_ready = (
    bool(
        nb11_identity_hits[
            "book_tiny_region_identity"
        ][
            "whole_region_under_100"
        ]
    )

    and

    (
        bool(
            nb11_identity_hits[
                "book_tiny_region_identity"
            ][
                "document_id"
            ]
        )

        or

        bool(
            nb11_identity_hits[
                "book_tiny_region_identity"
            ][
                "source_document"
            ]
        )
    )
)


transcript_long_identity_ready = all(
    bool(
        nb11_identity_hits[
            "transcript_long_identity"
        ][needle]
    )
    for needle
    in [
        "transcript_long_utterances",
        "document_id",
        "utterance",
        "token_count",
    ]
)


transcript_fallback_identity_ready = (
    bool(
        nb11_identity_hits[
            "transcript_fallback_identity"
        ][
            "transcript_token_fallback_fragment"
        ]
    )

    and

    (
        bool(
            nb11_identity_hits[
                "transcript_fallback_identity"
            ][
                "source_document"
            ]
        )

        or

        bool(
            nb11_identity_hits[
                "transcript_fallback_identity"
            ][
                "document_id"
            ]
        )
    )
)


dependency_resolution_plan[
    "book"
][
    "long_section"
] = (
    "REUSE_PRECOMPUTED_EVIDENCE"
    if book_section_identity_ready
    else
    "REMEASURE_IN_NB12"
)


dependency_resolution_plan[
    "book"
][
    "large_table_run"
] = (
    "REUSE_PRECOMPUTED_EVIDENCE"
    if book_table_identity_ready
    else
    "REMEASURE_IN_NB12"
)


dependency_resolution_plan[
    "book"
][
    "intrinsically_tiny_narrative_region"
] = (
    "REUSE_PRECOMPUTED_EVIDENCE"
    if book_tiny_identity_ready
    else
    "REMEASURE_IN_NB12"
)


dependency_resolution_plan[
    "book"
][
    "intrinsically_tiny_table_region"
] = (
    "REUSE_PRECOMPUTED_EVIDENCE"
    if book_tiny_identity_ready
    else
    "REMEASURE_IN_NB12"
)


dependency_resolution_plan[
    "transcript"
][
    "long_utterance_over_500"
] = (
    "REUSE_PRECOMPUTED_EVIDENCE"
    if transcript_long_identity_ready
    else
    "REMEASURE_IN_NB12"
)


dependency_resolution_plan[
    "transcript"
][
    "long_utterance_over_800"
] = (
    "REUSE_PRECOMPUTED_EVIDENCE"
    if transcript_long_identity_ready
    else
    "REMEASURE_IN_NB12"
)


dependency_resolution_plan[
    "transcript"
][
    "token_fallback_sentence_case"
] = (
    "REUSE_PRECOMPUTED_EVIDENCE"
    if transcript_fallback_identity_ready
    else
    "REMEASURE_IN_NB12"
)


# ============================================================
# 8) Validation
# ============================================================

all_relation_files_resolved_gate = all(
    path is not None

    for path
    in resolved_relation_files.values()
)


relation_schema_nonempty_gate = all(
    len(
        relation_schemas[
            relation_type
        ]
    )
    > 0

    for relation_type
    in relation_file_names
)


relation_identity_fields_found_gate = all(
    len(
        relation_identity_fields[
            relation_type
        ]
    )
    > 0

    for relation_type
    in relation_file_names
)


all_dependency_plans_defined_gate = all(
    plan
    in {
        "REUSE_PRECOMPUTED_EVIDENCE",
        "REMEASURE_IN_NB12",
        "INSUFFICIENT_IDENTITY_EVIDENCE",
    }

    for source_plan
    in dependency_resolution_plan.values()

    for plan
    in source_plan.values()
)


no_dependency_counts_mutated_gate = all(
    candidate_states[
        source_type
    ][
        stratum
    ][
        "candidate_count"
    ]
    is None

    for source_type, strata
    in DEPENDENCY_STRATA.items()

    for stratum
    in strata
)


sampling_not_started_gate = (
    GOLDEN_DATASET_CONTRACT_V1_DRAFT[
        "sampling_policy"
    ][
        "status"
    ]
    == "NOT_YET_DEFINED"
)


question_generation_not_started_gate = (
    GOLDEN_DATASET_CONTRACT_V1_DRAFT[
        "question_answer_policy"
    ][
        "question_generation_status"
    ]
    == "NOT_STARTED"
)


cell_15_gates = {
    "all_three_nb09_relation_files_resolved":
        all_relation_files_resolved_gate,

    "nb09_relation_schemas_nonempty":
        relation_schema_nonempty_gate,

    "nb09_relation_identity_fields_found":
        relation_identity_fields_found_gate,

    "all_8_dependency_resolution_plans_defined":
        all_dependency_plans_defined_gate,

    "dependency_counts_not_mutated":
        no_dependency_counts_mutated_gate,

    "sampling_not_started":
        sampling_not_started_gate,

    "question_generation_not_started":
        question_generation_not_started_gate,
}


cell_15_pass = all(
    cell_15_gates.values()
)


# ============================================================
# Output
# ============================================================

print(
    "=== NB09 RELATION ARTIFACT SCHEMAS ==="
)


for relation_type in [
    "exact",
    "near",
    "cross_source",
]:

    path = resolved_relation_files[
        relation_type
    ]

    print(
        f"\n[{relation_type.upper()}]"
    )

    print(
        "path:",
        (
            path.relative_to(
                repo_root
            )
            if path is not None
            else None
        )
    )

    print(
        "top-level schema:",
        sorted(
            relation_schemas[
                relation_type
            ]
        )
    )

    print(
        "identity-bearing fields:"
    )

    for field_path, count in (
        relation_identity_fields[
            relation_type
        ].most_common()
    ):

        print(
            "-",
            field_path,
            ":",
            count,
        )

    print(
        "representative identity-only examples:"
    )

    for example in (
        relation_examples[
            relation_type
        ]
    ):

        print(
            json.dumps(
                example,
                ensure_ascii=False,
                sort_keys=True,
            )
        )


print(
    "\n=== NB11 IDENTITY LINKAGE INSPECTION ==="
)


for group_name, needle_map in (
    nb11_identity_hits.items()
):

    print(
        f"\n[{group_name}]"
    )

    for needle, positions in (
        needle_map.items()
    ):

        print(
            "-",
            needle,
            "->",
            positions,
        )


print(
    "\n=== NB11 RELEVANT OUTPUT SIGNALS ==="
)

for dependency_name, positions in (
    dependency_output_hits.items()
):

    print(
        "-",
        dependency_name,
        "->",
        positions,
    )


print(
    "\n=== DEPENDENCY RESOLUTION PLAN ==="
)

for source_type, source_plan in (
    dependency_resolution_plan.items()
):

    print(
        f"\n[{source_type.upper()}]"
    )

    for stratum, plan in (
        source_plan.items()
    ):

        print(
            "-",
            stratum,
            ":",
            plan,
        )


print(
    "\n=== CODE CELL 15 GATES ==="
)

for name, passed in (
    cell_15_gates.items()
):

    print(
        name,
        ":",
        "PASS"
        if passed
        else "FAIL"
    )


print(
    "\nCODE CELL 15 GATE:",
    "PASS"
    if cell_15_pass
    else "FAIL"
)


assert cell_15_pass, (
    "Dependency evidence schema inspection failed. "
    "Do not load the tokenizer or generate dependency pools "
    "until identity linkage is verified."
)
```

<h3>Persisted Evidence Materialization Audit</h3>

<p>
Code Cell 15 نشان داد که برای هر هشت Dependency، منطق و Identity Linkage لازم در NB09/NB11 وجود دارد.
اما وجود Variable یا Logic در Notebook به‌تنهایی به معنی Persist شدن Candidateهای Record-level نیست.
</p>

<p>
در این مرحله بررسی می‌کنیم آیا Evidence موردنیاز مستقیماً از Artifactها و Reportهای ذخیره‌شده
قابل Materialize است، یا لازم است همان Measurement معتبر NB11 در NB12 دوباره اجرا شود.
</p>

<p>
این تمایز مهم است:
</p>

<ul>
<li><code>REUSE_PERSISTED_RECORD_LEVEL_EVIDENCE</code>: شناسه و Measurement واقعی Caseها در فایل ذخیره‌شده موجود است.</li>
<li><code>RECOMPUTE_FROM_FROZEN_NB11_LOGIC</code>: منطق معتبر و tokenizer فریز شده‌اند، اما Candidateهای Record-level Persist نشده‌اند.</li>
</ul>

<p>
در این Cell هیچ tokenizerی Load نمی‌شود و هیچ Candidate Pool ساخته نمی‌شود.
هدف فقط تعیین مسیر اجرای صحیح برای مرحله‌ی بعد است.
</p>

</div>

```python
# Code Cell 16

# ============================================================
# Persisted Evidence Materialization Audit
#
# Read-only.
# No tokenizer loading.
# No candidate generation.
# No artifact writes.
# ============================================================


# ============================================================
# 1) Resolve relevant persisted NB11 report/artifact files
# ============================================================

persisted_dependency_files = {
    "root_cause_analysis":
        required_paths[
            "root_cause_analysis"
        ],

    "structural_benchmark":
        required_paths[
            "structural_benchmark"
        ],

    "reconstruction_evidence":
        required_paths[
            "reconstruction_evidence"
        ],

    "decision_readiness":
        required_paths[
            "decision_readiness"
        ],

    "evaluation_freeze_manifest":
        required_paths[
            "evaluation_freeze_manifest"
        ],

    "final_selection_requirements":
        required_paths[
            "final_selection_requirements"
        ],
}


persisted_dependency_payloads = {}


for name, path in (
    persisted_dependency_files.items()
):

    assert path.is_file(), (
        f"Missing persisted dependency file: {path}"
    )

    with path.open(
        "r",
        encoding="utf-8",
    ) as f:

        persisted_dependency_payloads[
            name
        ] = json.load(
            f
        )


# ============================================================
# 2) Search persisted payloads for dependency-specific signals
# ============================================================

dependency_signal_terms = {

    "book_tiny": {
        "tiny",
        "under_100",
        "whole_region",
        "table_region",
        "narrative_region",
        "protected_region",
        "document_id",
    },

    "book_section": {
        "section",
        "over_800",
        "section_token",
        "heading_path",
        "document_id",
    },

    "book_table_run": {
        "table_run",
        "table",
        "run_token",
        "document_id",
    },

    "transcript_long": {
        "long_utterance",
        "utterance",
        "over_500",
        "over_800",
        "token_count",
        "document_id",
    },

    "transcript_fallback": {
        "fallback",
        "token_fallback",
        "sentence",
        "fragment",
        "document_id",
    },
}


persisted_signal_matches = {
    dependency_name: {}
    for dependency_name
    in dependency_signal_terms
}


for dependency_name, terms in (
    dependency_signal_terms.items()
):

    for payload_name, payload in (
        persisted_dependency_payloads.items()
    ):

        matches = (
            recursive_signal_matches_local(
                payload,
                terms,
                max_value_chars=180,
            )
        )

        persisted_signal_matches[
            dependency_name
        ][
            payload_name
        ] = matches


# ============================================================
# 3) Detect record-level identity-bearing objects conservatively
# ============================================================

def collect_dicts_recursive(
    obj,
    output=None,
):

    if output is None:
        output = []

    if isinstance(
        obj,
        dict,
    ):

        output.append(
            obj
        )

        for value in obj.values():

            collect_dicts_recursive(
                value,
                output,
            )

    elif isinstance(
        obj,
        list,
    ):

        for value in obj:

            collect_dicts_recursive(
                value,
                output,
            )

    return output


all_persisted_dicts = []


for payload_name, payload in (
    persisted_dependency_payloads.items()
):

    for obj in collect_dicts_recursive(
        payload
    ):

        all_persisted_dicts.append(
            (
                payload_name,
                obj,
            )
        )


def object_has_any_key(
    obj,
    keys,
):

    return any(
        key in obj
        for key
        in keys
    )


def object_has_identity(
    obj,
):

    return object_has_any_key(
        obj,
        {
            "document_id",
            "document_ids",
            "member_document_ids",
            "source_document_id",
            "source_document_ids",
            "physical_file_id",
            "section_key",
            "work_id",
            "edition_id",
            "paragraph_id",
            "utterance_id",
            "transcript_unit_id",
        },
    )


def object_has_token_measurement(
    obj,
):

    return object_has_any_key(
        obj,
        {
            "token_count",
            "token_counts",
            "section_token_count",
            "table_run_token_count",
            "utterance_token_count",
            "sentence_token_count",
            "whole_region_tokens",
        },
    )


record_level_candidates = {
    "book_tiny": [],
    "book_section": [],
    "book_table_run": [],
    "transcript_long": [],
    "transcript_fallback": [],
}


for payload_name, obj in (
    all_persisted_dicts
):

    if not object_has_identity(
        obj
    ):
        continue

    rendered_keys = " ".join(
        str(key).lower()
        for key in obj.keys()
    )

    rendered_values = " ".join(
        str(value).lower()
        for value in obj.values()
        if not isinstance(
            value,
            (
                dict,
                list,
            ),
        )
    )

    searchable = (
        rendered_keys
        + " "
        + rendered_values
    )

    # --------------------------------------------------------
    # Book tiny
    # --------------------------------------------------------

    if (
        "tiny" in searchable
        or
        "under_100" in searchable
        or
        "protected_region" in searchable
    ):

        record_level_candidates[
            "book_tiny"
        ].append(
            (
                payload_name,
                obj,
            )
        )

    # --------------------------------------------------------
    # Book section
    # --------------------------------------------------------

    if (
        "section" in searchable

        and

        (
            object_has_token_measurement(
                obj
            )
            or
            "over_800" in searchable
        )
    ):

        record_level_candidates[
            "book_section"
        ].append(
            (
                payload_name,
                obj,
            )
        )

    # --------------------------------------------------------
    # Book table run
    # --------------------------------------------------------

    if (
        "table" in searchable

        and

        "run" in searchable

        and

        (
            object_has_token_measurement(
                obj
            )
            or
            "over_500" in searchable
        )
    ):

        record_level_candidates[
            "book_table_run"
        ].append(
            (
                payload_name,
                obj,
            )
        )

    # --------------------------------------------------------
    # Transcript long utterance
    # --------------------------------------------------------

    if (
        "utterance" in searchable

        and

        (
            object_has_token_measurement(
                obj
            )
            or
            "over_500" in searchable
            or
            "over_800" in searchable
        )
    ):

        record_level_candidates[
            "transcript_long"
        ].append(
            (
                payload_name,
                obj,
            )
        )

    # --------------------------------------------------------
    # Transcript token fallback
    # --------------------------------------------------------

    if (
        (
            "fallback" in searchable
            or
            "token_offset" in searchable
        )

        and

        (
            "sentence" in searchable
            or
            "fragment" in searchable
        )
    ):

        record_level_candidates[
            "transcript_fallback"
        ].append(
            (
                payload_name,
                obj,
            )
        )


# ============================================================
# 4) Check whether persisted objects are sufficient for
#    exact Golden candidate reconstruction.
# ============================================================

persisted_record_level_readiness = {}


for dependency_name, items in (
    record_level_candidates.items()
):

    usable_items = []

    for payload_name, obj in items:

        if not object_has_identity(
            obj
        ):
            continue

        # For token-derived strata, an actual measurement
        # or explicit membership marker must be present.
        has_measurement = (
            object_has_token_measurement(
                obj
            )
        )

        searchable = (
            " ".join(
                str(key).lower()
                for key in obj.keys()
            )
            + " "
            + " ".join(
                str(value).lower()
                for value in obj.values()
                if not isinstance(
                    value,
                    (
                        dict,
                        list,
                    ),
                )
            )
        )

        has_explicit_membership_signal = any(
            term in searchable
            for term
            in [
                "under_100",
                "over_500",
                "over_800",
                "token_fallback",
                "fallback_fragment",
                "long_utterance",
            ]
        )

        if (
            has_measurement
            or
            has_explicit_membership_signal
        ):

            usable_items.append(
                (
                    payload_name,
                    obj,
                )
            )

    persisted_record_level_readiness[
        dependency_name
    ] = {
        "candidate_object_count":
            len(
                items
            ),

        "usable_record_level_count":
            len(
                usable_items
            ),

        "usable":
            len(
                usable_items
            )
            > 0,

        "examples":
            usable_items[
                :5
            ],
    }


# ============================================================
# 5) Website relation evidence is independently persisted
# ============================================================

website_relation_persisted_gate = all(
    resolved_relation_files[
        relation_type
    ]
    is not None

    and

    resolved_relation_files[
        relation_type
    ].is_file()

    for relation_type
    in [
        "exact",
        "near",
        "cross_source",
    ]
)


website_identity_persisted_gate = all(
    len(
        relation_identity_fields[
            relation_type
        ]
    )
    > 0

    for relation_type
    in [
        "exact",
        "near",
        "cross_source",
    ]
)


# ============================================================
# 6) Final execution plan
# ============================================================

dependency_materialization_plan = {

    "website": {
        "citation_locator_sensitive_case":
            (
                "REUSE_PERSISTED_RECORD_LEVEL_EVIDENCE"
                if (
                    website_relation_persisted_gate
                    and
                    website_identity_persisted_gate
                )
                else
                "BLOCKED"
            ),
    },

    "book": {
        "intrinsically_tiny_narrative_region":
            (
                "REUSE_PERSISTED_RECORD_LEVEL_EVIDENCE"
                if persisted_record_level_readiness[
                    "book_tiny"
                ][
                    "usable"
                ]
                else
                "RECOMPUTE_FROM_FROZEN_NB11_LOGIC"
            ),

        "intrinsically_tiny_table_region":
            (
                "REUSE_PERSISTED_RECORD_LEVEL_EVIDENCE"
                if persisted_record_level_readiness[
                    "book_tiny"
                ][
                    "usable"
                ]
                else
                "RECOMPUTE_FROM_FROZEN_NB11_LOGIC"
            ),

        "long_section":
            (
                "REUSE_PERSISTED_RECORD_LEVEL_EVIDENCE"
                if persisted_record_level_readiness[
                    "book_section"
                ][
                    "usable"
                ]
                else
                "RECOMPUTE_FROM_FROZEN_NB11_LOGIC"
            ),

        "large_table_run":
            (
                "REUSE_PERSISTED_RECORD_LEVEL_EVIDENCE"
                if persisted_record_level_readiness[
                    "book_table_run"
                ][
                    "usable"
                ]
                else
                "RECOMPUTE_FROM_FROZEN_NB11_LOGIC"
            ),
    },

    "transcript": {
        "long_utterance_over_500":
            (
                "REUSE_PERSISTED_RECORD_LEVEL_EVIDENCE"
                if persisted_record_level_readiness[
                    "transcript_long"
                ][
                    "usable"
                ]
                else
                "RECOMPUTE_FROM_FROZEN_NB11_LOGIC"
            ),

        "long_utterance_over_800":
            (
                "REUSE_PERSISTED_RECORD_LEVEL_EVIDENCE"
                if persisted_record_level_readiness[
                    "transcript_long"
                ][
                    "usable"
                ]
                else
                "RECOMPUTE_FROM_FROZEN_NB11_LOGIC"
            ),

        "token_fallback_sentence_case":
            (
                "REUSE_PERSISTED_RECORD_LEVEL_EVIDENCE"
                if persisted_record_level_readiness[
                    "transcript_fallback"
                ][
                    "usable"
                ]
                else
                "RECOMPUTE_FROM_FROZEN_NB11_LOGIC"
            ),
    },
}


# ============================================================
# 7) Validation
# ============================================================

all_8_materialization_plans_defined_gate = (
    sum(
        len(
            source_plan
        )
        for source_plan
        in dependency_materialization_plan.values()
    )
    == 8

    and

    all(
        plan
        in {
            "REUSE_PERSISTED_RECORD_LEVEL_EVIDENCE",
            "RECOMPUTE_FROM_FROZEN_NB11_LOGIC",
            "BLOCKED",
        }

        for source_plan
        in dependency_materialization_plan.values()

        for plan
        in source_plan.values()
    )
)


website_not_blocked_gate = (
    dependency_materialization_plan[
        "website"
    ][
        "citation_locator_sensitive_case"
    ]
    != "BLOCKED"
)


no_candidate_counts_changed_gate = all(
    candidate_states[
        source_type
    ][
        stratum
    ][
        "candidate_count"
    ]
    is None

    for source_type, strata
    in DEPENDENCY_STRATA.items()

    for stratum
    in strata
)


no_tokenizer_loaded_gate = (
    "MEASUREMENT_TOKENIZER"
    not in globals()
)


sampling_not_started_gate = (
    GOLDEN_DATASET_CONTRACT_V1_DRAFT[
        "sampling_policy"
    ][
        "status"
    ]
    == "NOT_YET_DEFINED"
)


question_generation_not_started_gate = (
    GOLDEN_DATASET_CONTRACT_V1_DRAFT[
        "question_answer_policy"
    ][
        "question_generation_status"
    ]
    == "NOT_STARTED"
)


cell_16_gates = {
    "all_8_materialization_plans_defined":
        all_8_materialization_plans_defined_gate,

    "website_relation_evidence_persisted":
        website_not_blocked_gate,

    "dependency_candidate_counts_unchanged":
        no_candidate_counts_changed_gate,

    "measurement_tokenizer_not_loaded":
        no_tokenizer_loaded_gate,

    "sampling_not_started":
        sampling_not_started_gate,

    "question_generation_not_started":
        question_generation_not_started_gate,
}


cell_16_pass = all(
    cell_16_gates.values()
)


# ============================================================
# Output
# ============================================================

print(
    "=== PERSISTED DEPENDENCY FILES ==="
)

for name, path in (
    persisted_dependency_files.items()
):

    print(
        "-",
        name,
        "->",
        path.relative_to(
            repo_root
        ),
    )


print(
    "\n=== PERSISTED RECORD-LEVEL READINESS ==="
)

for dependency_name, state in (
    persisted_record_level_readiness.items()
):

    print(
        "\n[",
        dependency_name,
        "]",
        sep="",
    )

    print(
        "candidate_object_count:",
        state[
            "candidate_object_count"
        ],
    )

    print(
        "usable_record_level_count:",
        state[
            "usable_record_level_count"
        ],
    )

    print(
        "usable:",
        state[
            "usable"
        ],
    )

    for payload_name, obj in (
        state[
            "examples"
        ]
    ):

        compact = {
            key: value
            for key, value
            in obj.items()
            if key
            in {
                "document_id",
                "document_ids",
                "member_document_ids",
                "source_document_id",
                "physical_file_id",
                "section_key",
                "work_id",
                "edition_id",
                "paragraph_id",
                "utterance_id",
                "transcript_unit_id",
                "token_count",
                "section_token_count",
                "table_run_token_count",
                "utterance_token_count",
                "sentence_token_count",
                "whole_region_tokens",
                "status",
                "reason",
                "type",
                "structural_type",
            }
        }

        print(
            "- source:",
            payload_name,
            "| object:",
            json.dumps(
                compact,
                ensure_ascii=False,
                sort_keys=True,
            ),
        )


print(
    "\n=== FINAL DEPENDENCY MATERIALIZATION PLAN ==="
)

for source_type, source_plan in (
    dependency_materialization_plan.items()
):

    print(
        f"\n[{source_type.upper()}]"
    )

    for stratum, plan in (
        source_plan.items()
    ):

        print(
            "-",
            stratum,
            ":",
            plan,
        )


print(
    "\n=== CODE CELL 16 GATES ==="
)

for name, passed in (
    cell_16_gates.items()
):

    print(
        name,
        ":",
        "PASS"
        if passed
        else "FAIL"
    )


print(
    "\nCODE CELL 16 GATE:",
    "PASS"
    if cell_16_pass
    else "FAIL"
)


assert cell_16_pass, (
    "Persisted evidence materialization audit failed. "
    "Do not load the measurement tokenizer or build dependency "
    "candidate pools until the persisted-vs-recompute plan is valid."
)
```

<li>Model identity برابر <code>intfloat/multilingual-e5-small</code> باشد.</li>
<li>Snapshot انتخاب‌شده در NB11 declared snapshots و Local Cache موجود باشد.</li>
<li>Tokenizer class همان خانواده‌ی ثبت‌شده در NB11 باشد.</li>
<li><code>add_special_tokens=False</code> برای Measurement رعایت شود.</li>
<li>Public tokenizer encode و backend tokenizer encode روی Calibration Textها سازگار باشند.</li>
<li>هیچ Candidate Pool یا Golden Sample در این مرحله ساخته نشود.</li>
</ul>

</div>

```python
# Code Cell 17

# ============================================================
# Frozen NB11 Measurement Tokenizer Resolution + Load
#
# No network.
# No candidate generation.
# No artifact writes.
# ============================================================


# ============================================================
# 1) Recover tokenizer-loading evidence from NB11 source
# ============================================================

tokenizer_source_positions = []

tokenizer_source_snippets = []


for position, source in enumerate(
    nb11_code_sources,
    start=1,
):

    lowered = source.lower()

    if (
        "autotokenizer" in lowered
        or
        "measurement_tokenizer" in lowered
        or
        "local_files_only" in lowered
    ):

        tokenizer_source_positions.append(
            position
        )

        tokenizer_source_snippets.append(
            {
                "position":
                    position,

                "source":
                    source,
            }
        )


# ============================================================
# 2) Search tokenizer-related NB11 source for declared
#    snapshot IDs that are also available locally.
# ============================================================

tokenizer_source_text = "\n".join(
    item[
        "source"
    ]
    for item
    in tokenizer_source_snippets
)


tokenizer_snapshot_mentions = sorted(
    snapshot_id
    for snapshot_id
    in declared_local_overlap
    if (
        snapshot_id
        in tokenizer_source_text
    )
)


# ============================================================
# 3) Inspect tokenizer-related NB11 outputs for snapshot IDs.
# ============================================================

tokenizer_output_text = "\n".join(
    nb11_relevant_outputs.get(
        position,
        ""
    )
    for position
    in tokenizer_source_positions
)


tokenizer_output_snapshot_mentions = sorted(
    snapshot_id
    for snapshot_id
    in declared_local_overlap
    if (
        snapshot_id
        in tokenizer_output_text
    )
)


# ============================================================
# 4) Recover explicit tokenizer snapshot path-like strings
#    from NB11 code.
# ============================================================

snapshot_path_pattern = re.compile(
    r'[^"\']*models--intfloat--multilingual-e5-small'
    r'[^"\']*snapshots'
    r'[^"\']*'
    r'([0-9a-f]{40})'
)


explicit_snapshot_path_mentions = sorted(
    set(
        match.group(1)
        for match
        in snapshot_path_pattern.finditer(
            tokenizer_source_text
        )
    )
)


explicit_snapshot_path_mentions = [
    snapshot_id
    for snapshot_id
    in explicit_snapshot_path_mentions
    if snapshot_id
    in local_snapshot_ids
]


# ============================================================
# 5) Resolve exact frozen tokenizer snapshot conservatively
# ============================================================

snapshot_evidence_sets = {
    "source_snapshot_mentions":
        set(
            tokenizer_snapshot_mentions
        ),

    "output_snapshot_mentions":
        set(
            tokenizer_output_snapshot_mentions
        ),

    "explicit_snapshot_path_mentions":
        set(
            explicit_snapshot_path_mentions
        ),
}


nonempty_snapshot_evidence_sets = [
    evidence_set
    for evidence_set
    in snapshot_evidence_sets.values()
    if evidence_set
]


if nonempty_snapshot_evidence_sets:

    resolved_snapshot_candidates = set.intersection(
        *nonempty_snapshot_evidence_sets
    )

else:

    resolved_snapshot_candidates = set()


# ------------------------------------------------------------
# Fallback:
# If NB11 did not print the literal SHA in tokenizer-specific
# source/output, inspect declared local snapshots for tokenizer
# configuration identity.
#
# This DOES NOT choose one yet; it only profiles them.
# ------------------------------------------------------------

local_snapshot_profiles = {}


for item in usable_local_snapshots:

    snapshot_id = item[
        "snapshot"
    ]

    snapshot_path = item[
        "path"
    ]

    tokenizer_config_path = (
        snapshot_path
        / "tokenizer_config.json"
    )

    tokenizer_config = {}

    if tokenizer_config_path.is_file():

        with tokenizer_config_path.open(
            "r",
            encoding="utf-8",
        ) as f:

            tokenizer_config = json.load(
                f
            )

    local_snapshot_profiles[
        snapshot_id
    ] = {
        "path":
            snapshot_path,

        "tokenizer_class":
            tokenizer_config.get(
                "tokenizer_class"
            ),

        "name_or_path":
            tokenizer_config.get(
                "name_or_path"
            ),

        "model_max_length":
            tokenizer_config.get(
                "model_max_length"
            ),

        "files":
            item[
                "tokenizer_assets"
            ],
    }


# ============================================================
# 6) If literal SHA resolution is ambiguous, compare tokenizer
#    file hashes across the usable NB11-declared snapshots.
#
# If tokenizer assets are byte-identical, they are equivalent
# for measurement; otherwise exact snapshot remains unresolved.
# ============================================================

def file_sha256(
    path,
):

    h = hashlib.sha256()

    with path.open(
        "rb"
    ) as f:

        while True:

            block = f.read(
                1024 * 1024
            )

            if not block:
                break

            h.update(
                block
            )

    return h.hexdigest()


tokenizer_asset_fingerprints = {}


for snapshot_id, profile in (
    local_snapshot_profiles.items()
):

    snapshot_path = profile[
        "path"
    ]

    asset_hashes = {}

    for filename in [
        "tokenizer.json",
        "tokenizer_config.json",
        "special_tokens_map.json",
        "sentencepiece.bpe.model",
    ]:

        path = (
            snapshot_path
            / filename
        )

        if path.is_file():

            asset_hashes[
                filename
            ] = file_sha256(
                path
            )

    tokenizer_asset_fingerprints[
        snapshot_id
    ] = asset_hashes


declared_usable_snapshot_ids = sorted(
    snapshot_id
    for snapshot_id
    in local_snapshot_profiles
    if snapshot_id
    in nb11_declared_snapshots
)


declared_usable_fingerprints = {
    snapshot_id:
        tokenizer_asset_fingerprints[
            snapshot_id
        ]

    for snapshot_id
    in declared_usable_snapshot_ids
}


fingerprint_groups = defaultdict(
    list
)


for snapshot_id, fingerprint in (
    declared_usable_fingerprints.items()
):

    fingerprint_key = json.dumps(
        fingerprint,
        sort_keys=True,
    )

    fingerprint_groups[
        fingerprint_key
    ].append(
        snapshot_id
    )


# ============================================================
# 7) Final resolution rule
# ============================================================

resolution_basis = None

resolved_tokenizer_snapshot_id = None


if (
    len(
        resolved_snapshot_candidates
    )
    == 1
):

    resolved_tokenizer_snapshot_id = next(
        iter(
            resolved_snapshot_candidates
        )
    )

    resolution_basis = (
        "NB11_TOKENIZER_SPECIFIC_SOURCE_OR_OUTPUT_EVIDENCE"
    )


elif (
    len(
        declared_usable_snapshot_ids
    )
    == 1
):

    resolved_tokenizer_snapshot_id = (
        declared_usable_snapshot_ids[
            0
        ]
    )

    resolution_basis = (
        "ONLY_ONE_DECLARED_USABLE_LOCAL_SNAPSHOT"
    )


elif (
    len(
        fingerprint_groups
    )
    == 1

    and

    len(
        declared_usable_snapshot_ids
    )
    > 0
):

    # All declared usable local snapshots have identical
    # tokenizer assets. Selection is deterministic only
    # because their tokenizer fingerprints are identical.

    resolved_tokenizer_snapshot_id = (
        declared_usable_snapshot_ids[
            0
        ]
    )

    resolution_basis = (
        "DECLARED_LOCAL_SNAPSHOTS_HAVE_IDENTICAL_TOKENIZER_ASSETS"
    )


resolved_tokenizer_snapshot_path = (
    local_snapshot_profiles[
        resolved_tokenizer_snapshot_id
    ][
        "path"
    ]

    if (
        resolved_tokenizer_snapshot_id
        is not None
    )

    else None
)


# ============================================================
# 8) Resolution gate BEFORE tokenizer load
# ============================================================

snapshot_resolved_gate = (
    resolved_tokenizer_snapshot_id
    is not None

    and

    resolved_tokenizer_snapshot_id
    in nb11_declared_snapshots

    and

    resolved_tokenizer_snapshot_id
    in local_snapshot_ids

    and

    resolved_tokenizer_snapshot_path.is_dir()
)


print(
    "=== TOKENIZER SNAPSHOT RESOLUTION ==="
)

print(
    "Tokenizer-related NB11 code cells:",
    tokenizer_source_positions
)

print(
    "Source SHA mentions:",
    tokenizer_snapshot_mentions
)

print(
    "Output SHA mentions:",
    tokenizer_output_snapshot_mentions
)

print(
    "Explicit path SHA mentions:",
    explicit_snapshot_path_mentions
)

print(
    "Declared usable local snapshots:",
    declared_usable_snapshot_ids
)

print(
    "Fingerprint groups:",
    {
        index + 1:
            snapshot_ids
        for index, snapshot_ids
        in enumerate(
            fingerprint_groups.values()
        )
    }
)

print(
    "Resolved snapshot:",
    resolved_tokenizer_snapshot_id
)

print(
    "Resolution basis:",
    resolution_basis
)

print(
    "Resolved path:",
    resolved_tokenizer_snapshot_path
)

print(
    "Snapshot resolution gate:",
    "PASS"
    if snapshot_resolved_gate
    else "FAIL"
)


assert snapshot_resolved_gate, (
    "Exact/equivalent frozen NB11 tokenizer snapshot "
    "could not be resolved without guessing. "
    "Do not load a tokenizer."
)


# ============================================================
# 9) Offline tokenizer load
# ============================================================

MEASUREMENT_TOKENIZER = (
    AutoTokenizer.from_pretrained(
        str(
            resolved_tokenizer_snapshot_path
        ),
        local_files_only=True,
        use_fast=True,
    )
)


# ============================================================
# 10) Frozen measurement wrappers
# ============================================================

def measurement_token_ids(
    text,
):

    return MEASUREMENT_TOKENIZER.encode(
        str(
            text
            or ""
        ),
        add_special_tokens=False,
    )


def token_count_text(
    text,
):

    return len(
        measurement_token_ids(
            text
        )
    )


# ============================================================
# 11) Backend/public encode equivalence calibration
# ============================================================

calibration_texts = [
    "",
    "سواد مالی",
    "پول چیست؟",
    "درآمد، پس‌انداز و سرمایه‌گذاری",
    "Financial literacy",
    "۱۲۳۴۵ 12345",
    "این یک جملهٔ آزمایشی است. جملهٔ دوم نیز اینجاست.",
]


tokenizer_calibration = []


backend_equivalence_gate = True


for text in calibration_texts:

    public_ids = (
        MEASUREMENT_TOKENIZER.encode(
            text,
            add_special_tokens=False,
        )
    )

    backend_ids = (
        MEASUREMENT_TOKENIZER.backend_tokenizer.encode(
            text,
            add_special_tokens=False,
        ).ids
    )

    equal = (
        public_ids
        == backend_ids
    )

    if not equal:

        backend_equivalence_gate = False

    tokenizer_calibration.append(
        {
            "text":
                text,

            "token_count":
                len(
                    public_ids
                ),

            "public_backend_equal":
                equal,
        }
    )


# ============================================================
# 12) Tokenizer identity validation
# ============================================================

tokenizer_class_name = (
    MEASUREMENT_TOKENIZER.__class__.__name__
)


tokenizer_model_max_length = (
    MEASUREMENT_TOKENIZER.model_max_length
)


tokenizer_identity_gate = (
    measurement_model_name
    ==
    expected_measurement_model
)


tokenizer_class_gate = (
    tokenizer_class_name
    ==
    "XLMRobertaTokenizerFast"
)


tokenizer_fast_gate = bool(
    getattr(
        MEASUREMENT_TOKENIZER,
        "is_fast",
        False,
    )
)


model_max_length_gate = (
    tokenizer_model_max_length
    == 512
)


measurement_policy_gate = (
    GOLDEN_DATASET_CONTRACT_V1_DRAFT[
        "token_measurement_policy"
    ][
        "model_name"
    ]
    ==
    expected_measurement_model

    and

    GOLDEN_DATASET_CONTRACT_V1_DRAFT[
        "token_measurement_policy"
    ][
        "add_special_tokens"
    ]
    is False
)


# ============================================================
# 13) Ensure dependency pools are still untouched
# ============================================================

dependency_counts_still_none_gate = all(
    candidate_states[
        source_type
    ][
        stratum
    ][
        "candidate_count"
    ]
    is None

    for source_type, strata
    in DEPENDENCY_STRATA.items()

    for stratum
    in strata
)


sampling_not_started_gate = (
    GOLDEN_DATASET_CONTRACT_V1_DRAFT[
        "sampling_policy"
    ][
        "status"
    ]
    == "NOT_YET_DEFINED"
)


question_generation_not_started_gate = (
    GOLDEN_DATASET_CONTRACT_V1_DRAFT[
        "question_answer_policy"
    ][
        "question_generation_status"
    ]
    == "NOT_STARTED"
)


cell_17_gates = {
    "frozen_snapshot_resolved_without_guessing":
        snapshot_resolved_gate,

    "measurement_model_identity":
        tokenizer_identity_gate,

    "expected_tokenizer_class":
        tokenizer_class_gate,

    "fast_tokenizer":
        tokenizer_fast_gate,

    "model_max_length_512":
        model_max_length_gate,

    "public_backend_encode_equivalent":
        backend_equivalence_gate,

    "add_special_tokens_false_policy":
        measurement_policy_gate,

    "dependency_candidate_counts_untouched":
        dependency_counts_still_none_gate,

    "sampling_not_started":
        sampling_not_started_gate,

    "question_generation_not_started":
        question_generation_not_started_gate,
}


cell_17_pass = all(
    cell_17_gates.values()
)


# ============================================================
# Output
# ============================================================

print(
    "\n=== LOADED MEASUREMENT TOKENIZER ==="
)

print(
    "Model:",
    measurement_model_name
)

print(
    "Snapshot:",
    resolved_tokenizer_snapshot_id
)

print(
    "Resolution basis:",
    resolution_basis
)

print(
    "Class:",
    tokenizer_class_name
)

print(
    "is_fast:",
    MEASUREMENT_TOKENIZER.is_fast
)

print(
    "model_max_length:",
    tokenizer_model_max_length
)

print(
    "add_special_tokens for measurement:",
    False
)


print(
    "\n=== TOKENIZER CALIBRATION ==="
)

for item in (
    tokenizer_calibration
):

    print(
        json.dumps(
            item,
            ensure_ascii=False,
            sort_keys=True,
        )
    )


print(
    "\n=== CODE CELL 17 GATES ==="
)

for name, passed in (
    cell_17_gates.items()
):

    print(
        name,
        ":",
        "PASS"
        if passed
        else "FAIL"
    )


print(
    "\nCODE CELL 17 GATE:",
    "PASS"
    if cell_17_pass
    else "FAIL"
)


assert cell_17_pass, (
    "Frozen measurement tokenizer validation failed. "
    "Do not recompute Book or Transcript dependency strata."
)
```

<div style="direction: rtl; white-space: normal; line-height: 1.7;">

<h3>Materialize Dependency Candidate Pools</h3>

<p>
در مراحل قبل مشخص شد که تنها Stratum وابسته‌ی Website می‌تواند مستقیماً از Evidence رکوردی NB09 ساخته شود،
در حالی که هفت Stratum مربوط به Book و Transcript باید با همان منطق Measurement فریز‌شده‌ی NB11 دوباره محاسبه شوند.
</p>

<p>
این مرحله هشت Dependency باقی‌مانده را Materialize می‌کند، اما هنوز Sampling انجام نمی‌دهد.
</p>

<h4>Website</h4>

<p>
برای <code>citation_locator_sensitive_case</code>، Exact/Near/Cross-source relationهای NB09
به سطح Website Document تجمیع می‌شوند.
وجود Relation فقط یک <strong>Review Prefilter</strong> است؛
Similarity یا Duplicate Relation به‌تنهایی اثبات نمی‌کند که Case نهایی Citation-sensitive است.
</p>

<h4>Book</h4>

<p>
منطق NB11 برای Book دقیقاً بازسازی می‌شود:
Paragraph و Supplemental Text از هم تفکیک می‌شوند،
Supplementalها به Parent خود متصل می‌شوند،
Sectionها بر اساس Source/Work/Edition/Heading Path ساخته می‌شوند
و سپس تغییر وضعیت Table/Narrative مرز Protected Region را تشکیل می‌دهد.
</p>

<ul>
<li><code>intrinsically_tiny_narrative_region</code>: Protected Narrative Region با کمتر از ۱۰۰ Token.</li>
<li><code>intrinsically_tiny_table_region</code>: Protected Table Region با کمتر از ۱۰۰ Token.</li>
<li><code>long_section</code>: Section با بیش از ۸۰۰ Token.</li>
<li><code>large_table_run</code>: Contiguous Table Run با بیش از ۵۰۰ Token.</li>
</ul>

<h4>Transcript</h4>

<p>
هر Utterance با Measurement Tokenizer فریز‌شده اندازه‌گیری می‌شود.
Sentence Segmentation نیز همان Regex فریز‌شده‌ی NB11 را استفاده می‌کند.
</p>

<ul>
<li><code>long_utterance_over_500</code>: Source Utterance با بیش از ۵۰۰ Token.</li>
<li><code>long_utterance_over_800</code>: Source Utterance با بیش از ۸۰۰ Token.</li>
<li><code>token_fallback_sentence_case</code>: Sentence منبعی که از Hard Maximum مربوط به Candidate عبور می‌کند؛ برای Stratum پایه، Sentenceهای بیش از ۵۰۰ Token ثبت می‌شوند و Applicability برای ۵۰۰ و ۸۰۰ جداگانه نگهداری می‌شود.</li>
</ul>

<p>
برای جلوگیری از Drift، Populationهای بازسازی‌شده با Facts اندازه‌گیری‌شده‌ی NB11
مانند ۱۸۳ Book Section، ۲۴۷ Protected Book Region، ۳۹ Table Run،
۵۱ Utterance بالای ۵۰۰ Token و ۲ Sentence بالای ۵۰۰ Token تطبیق داده می‌شوند.
</p>

<p>
هیچ Golden Sample، Question، Expected Answer یا Artifact در این مرحله ساخته یا نوشته نمی‌شود.
</p>

</div>

```python
# Code Cell 18

# ============================================================
# Materialize all 8 dependency candidate pools.
#
# Website:
#   Reuse persisted NB09 relation evidence.
#
# Book / Transcript:
#   Recompute from frozen NB11 logic using the validated
#   measurement tokenizer.
#
# No sampling.
# No question generation.
# No artifact writes.
# ============================================================


# ============================================================
# Frozen NB11-compatible helpers
# ============================================================

def nb11_normalized_heading_path(
    value,
):
    if isinstance(
        value,
        list,
    ):
        return tuple(
            str(item)
            for item in value
        )

    if value is None:
        return tuple()

    return (
        str(value),
    )


def nb11_token_count(
    text,
):
    encoding = (
        MEASUREMENT_TOKENIZER
        .backend_tokenizer
        .encode(
            str(
                text
                or ""
            ),
            add_special_tokens=False,
        )
    )

    return len(
        encoding.ids
    )


def nb11_sentence_spans_with_offsets(
    text,
):
    text = str(
        text
        or ""
    )

    if not text:
        return []

    separator_pattern = re.compile(
        r"(?<=[.!؟?!؛])\s+|\n+"
    )

    spans = []

    cursor = 0

    for match in (
        separator_pattern.finditer(
            text
        )
    ):
        raw_start = cursor
        raw_end = match.start()

        segment = text[
            raw_start:
            raw_end
        ]

        left_trim = (
            len(segment)
            - len(
                segment.lstrip()
            )
        )

        right_trim = (
            len(segment)
            - len(
                segment.rstrip()
            )
        )

        start = (
            raw_start
            + left_trim
        )

        end = (
            raw_end
            - right_trim
        )

        if end > start:

            spans.append(
                {
                    "start":
                        start,

                    "end":
                        end,

                    "text":
                        text[
                            start:end
                        ],
                }
            )

        cursor = match.end()

    tail = text[
        cursor:
    ]

    left_trim = (
        len(tail)
        - len(
            tail.lstrip()
        )
    )

    right_trim = (
        len(tail)
        - len(
            tail.rstrip()
        )
    )

    start = (
        cursor
        + left_trim
    )

    end = (
        len(text)
        - right_trim
    )

    if end > start:

        spans.append(
            {
                "start":
                    start,

                "end":
                    end,

                "text":
                    text[
                        start:end
                    ],
            }
        )

    return spans


# ============================================================
# 1) Build accepted-corpus lookup + exact NB11 atomic records
# ============================================================

accepted_lookup = {}

nb12_book_paragraph_records = []
nb12_book_supplemental_records = []
nb12_transcript_atomic_records = []

source_counts_recomputed = Counter()


with corpus_path.open(
    "r",
    encoding="utf-8",
) as handle:

    for line in handle:

        if not line.strip():
            continue

        record = json.loads(
            line
        )

        document_id = (
            record.get(
                "document_id"
            )
        )

        source_type = (
            record.get(
                "source_type"
            )
        )

        metadata = (
            record.get(
                "source_metadata"
            )
            or {}
        )

        locator = (
            record.get(
                "locator"
            )
            or {}
        )

        source_counts_recomputed[
            source_type
        ] += 1

        accepted_lookup[
            document_id
        ] = {
            "document_id":
                document_id,

            "source_id":
                record.get(
                    "source_id"
                ),

            "source_type":
                source_type,

            "locator":
                locator,

            "title":
                record.get(
                    "title"
                ),
        }

        # ----------------------------------------------------
        # BOOK — exact NB11 atomic structure
        # ----------------------------------------------------

        if source_type == "book":

            unit_type = (
                metadata.get(
                    "book_unit_type"
                )
            )

            common_book_item = {
                "document_id":
                    document_id,

                "source_id":
                    record.get(
                        "source_id"
                    ),

                "text":
                    str(
                        record.get(
                            "text"
                        )
                        or ""
                    ),

                "work_id":
                    (
                        metadata.get(
                            "work_id"
                        )
                        or locator.get(
                            "work_id"
                        )
                    ),

                "edition_id":
                    (
                        metadata.get(
                            "edition_id"
                        )
                        or locator.get(
                            "edition_id"
                        )
                    ),

                "paragraph_number":
                    metadata.get(
                        "paragraph_number"
                    ),

                "heading_path":
                    nb11_normalized_heading_path(
                        metadata.get(
                            "heading_path"
                        )
                    ),

                "is_heading":
                    metadata.get(
                        "is_heading"
                    ),

                "in_table":
                    metadata.get(
                        "in_table"
                    ),

                "parent_paragraph_id":
                    (
                        metadata.get(
                            "parent_paragraph_id"
                        )
                        or locator.get(
                            "parent_paragraph_id"
                        )
                    ),

                "paragraph_id":
                    locator.get(
                        "paragraph_id"
                    ),
            }

            if (
                unit_type
                == "paragraph"
            ):

                nb12_book_paragraph_records.append(
                    common_book_item
                )

            elif (
                unit_type
                == "supplemental_text"
            ):

                nb12_book_supplemental_records.append(
                    common_book_item
                )

        # ----------------------------------------------------
        # TRANSCRIPT — exact NB11 source-level fields
        # ----------------------------------------------------

        elif source_type == "transcript":

            speaker = (
                metadata.get(
                    "speaker"
                )
                or {}
            )

            base = (
                speaker.get(
                    "base"
                )
                or {}
            )

            effective = (
                speaker.get(
                    "effective"
                )
                or {}
            )

            nb12_transcript_atomic_records.append(
                {
                    "document_id":
                        document_id,

                    "source_id":
                        record.get(
                            "source_id"
                        ),

                    "text":
                        str(
                            record.get(
                                "text"
                            )
                            or ""
                        ),

                    "program_id":
                        (
                            metadata.get(
                                "program_id"
                            )
                            or locator.get(
                                "program_id"
                            )
                        ),

                    "physical_file_id":
                        locator.get(
                            "physical_file_id"
                        ),

                    "utterance_index":
                        metadata.get(
                            "utterance_index"
                        ),

                    "base_status":
                        base.get(
                            "status"
                        ),

                    "effective_label":
                        effective.get(
                            "label"
                        ),

                    "effective_canonical":
                        effective.get(
                            "canonical"
                        ),

                    "effective_primary":
                        effective.get(
                            "primary_dr_roudi_corpus"
                        ),
                }
            )


# ============================================================
# 2) WEBSITE — materialize citation-sensitive review prefilter
# ============================================================

website_relation_evidence = defaultdict(
    lambda: {
        "relation_ids":
            set(),

        "relation_types":
            set(),

        "relation_sources":
            set(),

        "neighbor_document_ids":
            set(),

        "neighbor_source_types":
            set(),
    }
)


def add_website_relation_evidence(
    website_document_id,
    relation_id,
    relation_type,
    relation_source,
    neighbor_document_ids,
):

    if (
        website_document_id
        not in accepted_lookup
    ):
        return

    if (
        accepted_lookup[
            website_document_id
        ][
            "source_type"
        ]
        != "website"
    ):
        return

    evidence = (
        website_relation_evidence[
            website_document_id
        ]
    )

    if relation_id is not None:
        evidence[
            "relation_ids"
        ].add(
            str(
                relation_id
            )
        )

    if relation_type is not None:
        evidence[
            "relation_types"
        ].add(
            str(
                relation_type
            )
        )

    evidence[
        "relation_sources"
    ].add(
        relation_source
    )

    for neighbor_id in (
        neighbor_document_ids
    ):

        if (
            neighbor_id
            == website_document_id
        ):
            continue

        evidence[
            "neighbor_document_ids"
        ].add(
            neighbor_id
        )

        neighbor_record = (
            accepted_lookup.get(
                neighbor_id
            )
        )

        if neighbor_record:

            evidence[
                "neighbor_source_types"
            ].add(
                neighbor_record[
                    "source_type"
                ]
            )


# ------------------------------------------------------------
# Exact relations
# ------------------------------------------------------------

with resolved_relation_files[
    "exact"
].open(
    "r",
    encoding="utf-8",
) as handle:

    for line in handle:

        if not line.strip():
            continue

        relation = json.loads(
            line
        )

        members = [
            document_id
            for document_id
            in (
                relation.get(
                    "member_document_ids"
                )
                or []
            )
            if document_id
            in accepted_lookup
        ]

        website_members = [
            document_id
            for document_id
            in members
            if (
                accepted_lookup[
                    document_id
                ][
                    "source_type"
                ]
                == "website"
            )
        ]

        for website_document_id in (
            website_members
        ):

            add_website_relation_evidence(
                website_document_id=(
                    website_document_id
                ),
                relation_id=(
                    relation.get(
                        "relation_group_id"
                    )
                ),
                relation_type=(
                    relation.get(
                        "relation_type"
                    )
                ),
                relation_source=(
                    "NB09_EXACT"
                ),
                neighbor_document_ids=(
                    members
                ),
            )


# ------------------------------------------------------------
# Near + Cross-source relations
# ------------------------------------------------------------

for relation_source, relation_key in [
    (
        "NB09_NEAR",
        "near",
    ),
    (
        "NB09_CROSS_SOURCE",
        "cross_source",
    ),
]:

    with resolved_relation_files[
        relation_key
    ].open(
        "r",
        encoding="utf-8",
    ) as handle:

        for line in handle:

            if not line.strip():
                continue

            relation = json.loads(
                line
            )

            document_id_a = (
                relation.get(
                    "document_id_a"
                )
            )

            document_id_b = (
                relation.get(
                    "document_id_b"
                )
            )

            pair = [
                document_id_a,
                document_id_b,
            ]

            for website_document_id, neighbor_id in [
                (
                    document_id_a,
                    document_id_b,
                ),
                (
                    document_id_b,
                    document_id_a,
                ),
            ]:

                if (
                    website_document_id
                    not in accepted_lookup
                ):
                    continue

                if (
                    accepted_lookup[
                        website_document_id
                    ][
                        "source_type"
                    ]
                    != "website"
                ):
                    continue

                add_website_relation_evidence(
                    website_document_id=(
                        website_document_id
                    ),
                    relation_id=(
                        relation.get(
                            "relation_id"
                        )
                    ),
                    relation_type=(
                        relation.get(
                            "relation_type"
                        )
                    ),
                    relation_source=(
                        relation_source
                    ),
                    neighbor_document_ids=(
                        pair
                    ),
                )


candidate_pools[
    "website"
][
    "citation_locator_sensitive_case"
] = []


for document_id in sorted(
    website_relation_evidence
):

    evidence = (
        website_relation_evidence[
            document_id
        ]
    )

    if not evidence[
        "neighbor_document_ids"
    ]:
        continue

    source_record = (
        accepted_lookup[
            document_id
        ]
    )

    locator = (
        source_record[
            "locator"
        ]
        or {}
    )

    candidate_pools[
        "website"
    ][
        "citation_locator_sensitive_case"
    ].append(
        {
            "document_id":
                document_id,

            "source_id":
                source_record[
                    "source_id"
                ],

            "canonical_url":
                locator.get(
                    "canonical_url"
                ),

            "url":
                locator.get(
                    "url"
                ),

            "relation_count":
                len(
                    evidence[
                        "relation_ids"
                    ]
                ),

            "relation_ids":
                sorted(
                    evidence[
                        "relation_ids"
                    ]
                ),

            "relation_types":
                sorted(
                    evidence[
                        "relation_types"
                    ]
                ),

            "relation_sources":
                sorted(
                    evidence[
                        "relation_sources"
                    ]
                ),

            "neighbor_document_ids":
                sorted(
                    evidence[
                        "neighbor_document_ids"
                    ]
                ),

            "neighbor_source_types":
                sorted(
                    evidence[
                        "neighbor_source_types"
                    ]
                ),

            "review_status":
                "HUMAN_REVIEW_REQUIRED",

            "citation_sensitive_confirmed":
                False,
        }
    )


candidate_states[
    "website"
][
    "citation_locator_sensitive_case"
] = {
    "status":
        "REVIEW_PREFILTER_READY",

    "candidate_count":
        len(
            candidate_pools[
                "website"
            ][
                "citation_locator_sensitive_case"
            ]
        ),

    "final_membership_confirmed":
        False,

    "evidence_basis":
        "PERSISTED_NB09_RELATION_EVIDENCE",
}


# ============================================================
# 3) BOOK — exact NB11 section reconstruction
# ============================================================

nb12_book_sections = defaultdict(
    list
)


for item in (
    nb12_book_paragraph_records
):

    section_key = (
        str(
            item[
                "source_id"
            ]
        ),

        str(
            item[
                "work_id"
            ]
        ),

        str(
            item[
                "edition_id"
            ]
        ),

        item[
            "heading_path"
        ],
    )

    nb12_book_sections[
        section_key
    ].append(
        item
    )


nb12_book_section_stats = {}


for section_key, records in (
    nb12_book_sections.items()
):

    ordered = sorted(
        records,
        key=lambda item:
            item[
                "paragraph_number"
            ],
    )

    joined_text = "\n".join(
        item[
            "text"
        ]
        for item
        in ordered
    )

    section_tokens = (
        nb11_token_count(
            joined_text
        )
    )

    nb12_book_section_stats[
        section_key
    ] = {
        "token_count":
            section_tokens,

        "record_count":
            len(
                ordered
            ),

        "document_ids":
            [
                item[
                    "document_id"
                ]
                for item
                in ordered
            ],

        "start_paragraph":
            ordered[
                0
            ][
                "paragraph_number"
            ],

        "end_paragraph":
            ordered[
                -1
            ][
                "paragraph_number"
            ],

        "heading_path":
            section_key[
                3
            ],

        "table_record_count":
            sum(
                1
                for item
                in ordered
                if (
                    item[
                        "in_table"
                    ]
                    is True
                )
            ),
    }


candidate_pools[
    "book"
][
    "long_section"
] = []


for section_key, stats in sorted(
    nb12_book_section_stats.items(),
    key=lambda item:
        str(
            item[
                0
            ]
        ),
):

    if (
        stats[
            "token_count"
        ]
        <= 800
    ):
        continue

    candidate_pools[
        "book"
    ][
        "long_section"
    ].append(
        {
            "source_id":
                section_key[
                    0
                ],

            "work_id":
                section_key[
                    1
                ],

            "edition_id":
                section_key[
                    2
                ],

            "heading_path":
                list(
                    section_key[
                        3
                    ]
                ),

            "token_count":
                stats[
                    "token_count"
                ],

            "record_count":
                stats[
                    "record_count"
                ],

            "start_paragraph":
                stats[
                    "start_paragraph"
                ],

            "end_paragraph":
                stats[
                    "end_paragraph"
                ],

            "document_ids":
                stats[
                    "document_ids"
                ],

            "table_record_count":
                stats[
                    "table_record_count"
                ],
        }
    )


# ============================================================
# 4) BOOK — protected region reconstruction
# ============================================================

supplemental_by_parent = defaultdict(
    list
)


for supplemental in (
    nb12_book_supplemental_records
):

    parent_key = str(
        supplemental[
            "parent_paragraph_id"
        ]
    )

    supplemental_by_parent[
        parent_key
    ].append(
        supplemental
    )


nb12_book_protected_regions = []


for section_key, records in sorted(
    nb12_book_sections.items(),
    key=lambda pair:
        str(
            pair[
                0
            ]
        ),
):

    ordered = sorted(
        records,
        key=lambda item:
            item[
                "paragraph_number"
            ],
    )

    segment_index = 0
    current_table_flag = None
    current_units = []

    def flush_nb12_book_region():

        nonlocal_values = None

        if not current_units:
            return

        # region identity matches NB11 segmentation logic
        region_type = (
            "table"
            if current_table_flag
            else "narrative"
        )

        atomic_texts = [
            unit[
                "text"
            ]
            for unit
            in current_units
        ]

        region_content = "\n".join(
            atomic_texts
        )

        member_document_ids = [
            document_id
            for unit
            in current_units
            for document_id
            in unit[
                "member_document_ids"
            ]
        ]

        nb12_book_protected_regions.append(
            {
                "source_id":
                    section_key[
                        0
                    ],

                "work_id":
                    section_key[
                        1
                    ],

                "edition_id":
                    section_key[
                        2
                    ],

                "heading_path":
                    list(
                        section_key[
                            3
                        ]
                    ),

                "segment_index":
                    segment_index,

                "region_type":
                    region_type,

                "token_count":
                    nb11_token_count(
                        region_content
                    ),

                "atomic_unit_count":
                    len(
                        current_units
                    ),

                "member_document_ids":
                    member_document_ids,

                "start_paragraph":
                    current_units[
                        0
                    ][
                        "paragraph_number"
                    ],

                "end_paragraph":
                    current_units[
                        -1
                    ][
                        "paragraph_number"
                    ],
            }
        )

    for item in ordered:

        table_flag = (
            item[
                "in_table"
            ]
            is True
        )

        if (
            current_table_flag
            is None
        ):

            current_table_flag = (
                table_flag
            )

        elif (
            table_flag
            != current_table_flag
        ):

            segment_index += 1

            flush_nb12_book_region()

            current_units = []

            current_table_flag = (
                table_flag
            )

        parent_identity_candidates = [
            item[
                "document_id"
            ],
            item[
                "paragraph_id"
            ],
        ]

        supplements = []
        existing_supplement_ids = set()

        for parent_identity in (
            parent_identity_candidates
        ):

            if not parent_identity:
                continue

            for match in (
                supplemental_by_parent.get(
                    str(
                        parent_identity
                    ),
                    [],
                )
            ):

                supplement_id = (
                    match[
                        "document_id"
                    ]
                )

                if (
                    supplement_id
                    in existing_supplement_ids
                ):
                    continue

                supplements.append(
                    match
                )

                existing_supplement_ids.add(
                    supplement_id
                )

        member_ids = [
            item[
                "document_id"
            ],
            *[
                supplemental[
                    "document_id"
                ]
                for supplemental
                in supplements
            ],
        ]

        atomic_text = "\n".join(
            [
                item[
                    "text"
                ],
                *[
                    supplemental[
                        "text"
                    ]
                    for supplemental
                    in supplements
                ],
            ]
        )

        current_units.append(
            {
                "text":
                    atomic_text,

                "paragraph_number":
                    item[
                        "paragraph_number"
                    ],

                "member_document_ids":
                    member_ids,
            }
        )

    if current_units:

        segment_index += 1

        flush_nb12_book_region()


# ============================================================
# 5) BOOK — intrinsic tiny narrative/table regions
# ============================================================

candidate_pools[
    "book"
][
    "intrinsically_tiny_narrative_region"
] = []


candidate_pools[
    "book"
][
    "intrinsically_tiny_table_region"
] = []


for region in (
    nb12_book_protected_regions
):

    if (
        region[
            "token_count"
        ]
        >= 100
    ):
        continue

    target_stratum = (
        "intrinsically_tiny_table_region"
        if (
            region[
                "region_type"
            ]
            == "table"
        )
        else
        "intrinsically_tiny_narrative_region"
    )

    candidate_pools[
        "book"
    ][
        target_stratum
    ].append(
        dict(
            region
        )
    )


# ============================================================
# 6) BOOK — contiguous table runs (exact NB11 logic)
# ============================================================

book_records_by_edition = defaultdict(
    list
)


for item in (
    nb12_book_paragraph_records
):

    edition_key = (
        str(
            item[
                "work_id"
            ]
        ),

        str(
            item[
                "edition_id"
            ]
        ),
    )

    book_records_by_edition[
        edition_key
    ].append(
        item
    )


nb12_book_table_runs = []


for edition_key, records in (
    book_records_by_edition.items()
):

    ordered = sorted(
        records,
        key=lambda item:
            item[
                "paragraph_number"
            ],
    )

    current_run = []

    def flush_nb12_table_run():

        if not current_run:
            return

        joined_text = "\n".join(
            item[
                "text"
            ]
            for item
            in current_run
        )

        nb12_book_table_runs.append(
            {
                "work_id":
                    edition_key[
                        0
                    ],

                "edition_id":
                    edition_key[
                        1
                    ],

                "record_count":
                    len(
                        current_run
                    ),

                "token_count":
                    nb11_token_count(
                        joined_text
                    ),

                "start_paragraph":
                    current_run[
                        0
                    ][
                        "paragraph_number"
                    ],

                "end_paragraph":
                    current_run[
                        -1
                    ][
                        "paragraph_number"
                    ],

                "heading_paths":
                    sorted(
                        {
                            item[
                                "heading_path"
                            ]
                            for item
                            in current_run
                        },
                        key=str,
                    ),

                "document_ids":
                    [
                        item[
                            "document_id"
                        ]
                        for item
                        in current_run
                    ],
            }
        )

    for item in ordered:

        if (
            item[
                "in_table"
            ]
            is True
        ):

            current_run.append(
                item
            )

        else:

            flush_nb12_table_run()

            current_run = []

    flush_nb12_table_run()


candidate_pools[
    "book"
][
    "large_table_run"
] = [
    {
        **run,
        "heading_paths":
            [
                list(
                    path
                )
                for path
                in run[
                    "heading_paths"
                ]
            ],
    }

    for run in nb12_book_table_runs

    if (
        run[
            "token_count"
        ]
        > 500
    )
]


# ============================================================
# 7) TRANSCRIPT — long utterances + exact sentence fallback
# ============================================================

nb12_transcript_long_over_500 = []
nb12_transcript_long_over_800 = []
nb12_token_fallback_sentences = []


for item in (
    nb12_transcript_atomic_records
):

    text = (
        item[
            "text"
        ]
    )

    utterance_token_count = (
        nb11_token_count(
            text
        )
    )

    if (
        utterance_token_count
        > 500
    ):

        sentence_spans = (
            nb11_sentence_spans_with_offsets(
                text
            )
        )

        sentence_measurements = []

        for sentence_index, span in enumerate(
            sentence_spans,
            start=1,
        ):

            sentence_token_count = (
                nb11_token_count(
                    span[
                        "text"
                    ]
                )
            )

            sentence_measurements.append(
                {
                    "sentence_index":
                        sentence_index,

                    "char_start":
                        span[
                            "start"
                        ],

                    "char_end":
                        span[
                            "end"
                        ],

                    "token_count":
                        sentence_token_count,
                }
            )

            if (
                sentence_token_count
                > 500
            ):

                nb12_token_fallback_sentences.append(
                    {
                        "document_id":
                            item[
                                "document_id"
                            ],

                        "source_id":
                            item[
                                "source_id"
                            ],

                        "program_id":
                            item[
                                "program_id"
                            ],

                        "physical_file_id":
                            item[
                                "physical_file_id"
                            ],

                        "utterance_index":
                            item[
                                "utterance_index"
                            ],

                        "utterance_token_count":
                            utterance_token_count,

                        "sentence_index":
                            sentence_index,

                        "char_start":
                            span[
                                "start"
                            ],

                        "char_end":
                            span[
                                "end"
                            ],

                        "sentence_token_count":
                            sentence_token_count,

                        "fallback_for_hard_max_500":
                            True,

                        "fallback_for_hard_max_800":
                            (
                                sentence_token_count
                                > 800
                            ),

                        "expected_fragment_marker":
                            "transcript_token_fallback_fragment",

                        "base_status":
                            item[
                                "base_status"
                            ],

                        "effective_primary":
                            item[
                                "effective_primary"
                            ],
                    }
                )

        long_record = {
            "document_id":
                item[
                    "document_id"
                ],

            "source_id":
                item[
                    "source_id"
                ],

            "program_id":
                item[
                    "program_id"
                ],

            "physical_file_id":
                item[
                    "physical_file_id"
                ],

            "utterance_index":
                item[
                    "utterance_index"
                ],

            "token_count":
                utterance_token_count,

            "sentence_count":
                len(
                    sentence_measurements
                ),

            "max_sentence_tokens":
                (
                    max(
                        measurement[
                            "token_count"
                        ]
                        for measurement
                        in sentence_measurements
                    )
                    if sentence_measurements
                    else 0
                ),

            "base_status":
                item[
                    "base_status"
                ],

            "effective_primary":
                item[
                    "effective_primary"
                ],
        }

        nb12_transcript_long_over_500.append(
            long_record
        )

        if (
            utterance_token_count
            > 800
        ):

            nb12_transcript_long_over_800.append(
                dict(
                    long_record
                )
            )


candidate_pools[
    "transcript"
][
    "long_utterance_over_500"
] = (
    nb12_transcript_long_over_500
)


candidate_pools[
    "transcript"
][
    "long_utterance_over_800"
] = (
    nb12_transcript_long_over_800
)


candidate_pools[
    "transcript"
][
    "token_fallback_sentence_case"
] = (
    nb12_token_fallback_sentences
)


# ============================================================
# 8) Update candidate states for all 8 dependencies
# ============================================================

for stratum in [
    "intrinsically_tiny_narrative_region",
    "intrinsically_tiny_table_region",
    "long_section",
    "large_table_run",
]:

    candidate_states[
        "book"
    ][
        stratum
    ] = {
        "status":
            "EXACT_POOL_READY",

        "candidate_count":
            len(
                candidate_pools[
                    "book"
                ][
                    stratum
                ]
            ),

        "evidence_basis":
            "RECOMPUTED_FROM_FROZEN_NB11_LOGIC",
    }


for stratum in [
    "long_utterance_over_500",
    "long_utterance_over_800",
    "token_fallback_sentence_case",
]:

    candidate_states[
        "transcript"
    ][
        stratum
    ] = {
        "status":
            "EXACT_POOL_READY",

        "candidate_count":
            len(
                candidate_pools[
                    "transcript"
                ][
                    stratum
                ]
            ),

        "evidence_basis":
            "RECOMPUTED_FROM_FROZEN_NB11_LOGIC",
    }


# ============================================================
# 9) Update operational rule states
# ============================================================

for stratum in [
    "intrinsically_tiny_narrative_region",
    "intrinsically_tiny_table_region",
    "long_section",
    "large_table_run",
]:

    GOLDEN_DATASET_CONTRACT_V1_DRAFT[
        "stratum_rule_state"
    ][
        "book"
    ][
        stratum
    ] = (
        "OPERATIONAL_RULE_VALIDATED"
    )


for stratum in [
    "long_utterance_over_500",
    "long_utterance_over_800",
    "token_fallback_sentence_case",
]:

    GOLDEN_DATASET_CONTRACT_V1_DRAFT[
        "stratum_rule_state"
    ][
        "transcript"
    ][
        stratum
    ] = (
        "OPERATIONAL_RULE_VALIDATED"
    )


GOLDEN_DATASET_CONTRACT_V1_DRAFT[
    "stratum_rule_state"
][
    "website"
][
    "citation_locator_sensitive_case"
] = (
    "REVIEW_PREFILTER_VALIDATED"
)


# ============================================================
# 10) Independent parity checks against measured NB11 facts
# ============================================================

book_tiny_total = (
    len(
        candidate_pools[
            "book"
        ][
            "intrinsically_tiny_narrative_region"
        ]
    )
    +
    len(
        candidate_pools[
            "book"
        ][
            "intrinsically_tiny_table_region"
        ]
    )
)


table_runs_crossing_heading_paths = sum(
    1
    for run
    in nb12_book_table_runs
    if (
        len(
            run[
                "heading_paths"
            ]
        )
        > 1
    )
)


fallback_over_800_count = sum(
    1
    for item
    in nb12_token_fallback_sentences
    if (
        item[
            "fallback_for_hard_max_800"
        ]
    )
)


dependency_materialization_gates = {

    "accepted_source_counts_match":
        (
            source_counts_recomputed[
                "website"
            ]
            == 529

            and

            source_counts_recomputed[
                "book"
            ]
            == 2701

            and

            source_counts_recomputed[
                "transcript"
            ]
            == 16226
        ),

    "book_atomic_population_matches_nb11":
        (
            len(
                nb12_book_paragraph_records
            )
            == 2689

            and

            len(
                nb12_book_supplemental_records
            )
            == 12
        ),

    "book_sections_match_nb11_183":
        (
            len(
                nb12_book_sections
            )
            == 183
        ),

    "book_long_sections_over_800_match_nb11_14":
        (
            len(
                candidate_pools[
                    "book"
                ][
                    "long_section"
                ]
            )
            == 14
        ),

    "book_protected_regions_match_nb11_247":
        (
            len(
                nb12_book_protected_regions
            )
            == 247
        ),

    "book_intrinsic_tiny_total_matches_nb11_104":
        (
            book_tiny_total
            == 104
        ),

    "book_table_runs_match_nb11_39":
        (
            len(
                nb12_book_table_runs
            )
            == 39
        ),

    "book_table_runs_cross_heading_zero":
        (
            table_runs_crossing_heading_paths
            == 0
        ),

    "large_table_run_pool_nonempty":
        (
            len(
                candidate_pools[
                    "book"
                ][
                    "large_table_run"
                ]
            )
            > 0
        ),

    "transcript_population_matches_nb11":
        (
            len(
                nb12_transcript_atomic_records
            )
            == 16226
        ),

    "long_utterance_over_500_matches_nb11_51":
        (
            len(
                nb12_transcript_long_over_500
            )
            == 51
        ),

    "long_utterance_over_800_matches_nb11_15":
        (
            len(
                nb12_transcript_long_over_800
            )
            == 15
        ),

    "fallback_sentence_over_500_matches_nb11_2":
        (
            len(
                nb12_token_fallback_sentences
            )
            == 2
        ),

    "website_citation_review_prefilter_nonempty":
        (
            len(
                candidate_pools[
                    "website"
                ][
                    "citation_locator_sensitive_case"
                ]
            )
            > 0
        ),

    "website_citation_cases_not_auto_confirmed":
        all(
            item[
                "citation_sensitive_confirmed"
            ]
            is False

            and

            item[
                "review_status"
            ]
            == "HUMAN_REVIEW_REQUIRED"

            for item
            in candidate_pools[
                "website"
            ][
                "citation_locator_sensitive_case"
            ]
        ),

    "all_8_dependencies_materialized":
        all(
            candidate_states[
                source_type
            ][
                stratum
            ][
                "candidate_count"
            ]
            is not None

            for source_type, strata
            in DEPENDENCY_STRATA.items()

            for stratum
            in strata
        ),

    "sampling_not_started":
        (
            GOLDEN_DATASET_CONTRACT_V1_DRAFT[
                "sampling_policy"
            ][
                "status"
            ]
            == "NOT_YET_DEFINED"
        ),

    "question_generation_not_started":
        (
            GOLDEN_DATASET_CONTRACT_V1_DRAFT[
                "question_answer_policy"
            ][
                "question_generation_status"
            ]
            == "NOT_STARTED"
        ),
}


dependency_materialization_pass = all(
    dependency_materialization_gates.values()
)


# ============================================================
# 11) Output
# ============================================================

print(
    "=== DEPENDENCY MATERIALIZATION RESULTS ==="
)


print(
    "\n[WEBSITE]"
)

print(
    "- citation_locator_sensitive_case:",
    len(
        candidate_pools[
            "website"
        ][
            "citation_locator_sensitive_case"
        ]
    ),
    "| REVIEW_PREFILTER_READY"
)


print(
    "\n[BOOK]"
)

for stratum in [
    "intrinsically_tiny_narrative_region",
    "intrinsically_tiny_table_region",
    "long_section",
    "large_table_run",
]:

    print(
        "-",
        stratum,
        ":",
        len(
            candidate_pools[
                "book"
            ][
                stratum
            ]
        ),
    )


print(
    "\nBook structural parity:"
)

print(
    "- sections:",
    len(
        nb12_book_sections
    )
)

print(
    "- protected regions:",
    len(
        nb12_book_protected_regions
    )
)

print(
    "- intrinsic tiny total:",
    book_tiny_total
)

print(
    "- contiguous table runs:",
    len(
        nb12_book_table_runs
    )
)

print(
    "- table runs crossing heading paths:",
    table_runs_crossing_heading_paths
)


print(
    "\n[TRANSCRIPT]"
)

print(
    "- long_utterance_over_500:",
    len(
        nb12_transcript_long_over_500
    )
)

print(
    "- long_utterance_over_800:",
    len(
        nb12_transcript_long_over_800
    )
)

print(
    "- token_fallback_sentence_case:",
    len(
        nb12_token_fallback_sentences
    )
)

print(
    "- fallback sentences also >800:",
    fallback_over_800_count
)


print(
    "\n=== REPRESENTATIVE DEPENDENCY CANDIDATES ==="
)


for source_type, strata in [
    (
        "website",
        [
            "citation_locator_sensitive_case",
        ],
    ),
    (
        "book",
        [
            "intrinsically_tiny_narrative_region",
            "intrinsically_tiny_table_region",
            "long_section",
            "large_table_run",
        ],
    ),
    (
        "transcript",
        [
            "long_utterance_over_500",
            "long_utterance_over_800",
            "token_fallback_sentence_case",
        ],
    ),
]:

    print(
        f"\n[{source_type.upper()}]"
    )

    for stratum in strata:

        pool = candidate_pools[
            source_type
        ][
            stratum
        ]

        print(
            "\n",
            stratum,
            "| count:",
            len(
                pool
            ),
            sep="",
        )

        for example in pool[:2]:

            compact = {
                key:
                    (
                        value[:5]
                        if (
                            isinstance(
                                value,
                                list,
                            )
                            and len(value) > 5
                        )
                        else value
                    )

                for key, value
                in example.items()

                if key
                not in {
                    "relation_ids",
                    "neighbor_document_ids",
                    "document_ids",
                    "member_document_ids",
                }
            }

            for long_list_key in [
                "relation_ids",
                "neighbor_document_ids",
                "document_ids",
                "member_document_ids",
            ]:

                if (
                    long_list_key
                    in example
                ):

                    compact[
                        long_list_key
                        + "_count"
                    ] = len(
                        example[
                            long_list_key
                        ]
                    )

                    compact[
                        long_list_key
                        + "_preview"
                    ] = example[
                        long_list_key
                    ][:3]

            print(
                json.dumps(
                    compact,
                    ensure_ascii=False,
                    sort_keys=True,
                )
            )


print(
    "\n=== DEPENDENCY MATERIALIZATION GATES ==="
)

for name, passed in (
    dependency_materialization_gates.items()
):

    print(
        name,
        ":",
        "PASS"
        if passed
        else "FAIL"
    )


print(
    "\nCODE CELL 18 GATE:",
    "PASS"
    if dependency_materialization_pass
    else "FAIL"
)


assert dependency_materialization_pass, (
    "Dependency materialization failed. "
    "Do not define Golden sampling until the recomputed "
    "NB11 populations and persisted NB09 evidence pass parity."
)
```

<h3>Consolidated Golden Candidate Universe and Coverage Audit</h3>

<p>
پس از Materialize شدن همه‌ی Dependencyها، اکنون تمام ۲۷ Stratum اجباری NB12
دارای Candidate Population واقعی هستند.
قبل از تعریف Sampling Policy باید این Populationها در یک Registry واحد تجمیع و مستقل Validate شوند.
</p>

<p>
این مرحله برای هر Stratum موارد زیر را ثبت می‌کند:
</p>

<ul>
<li>نوع Source و Unit ارزیابی؛</li>
<li>Operational Eligibility؛</li>
<li>وضعیت Candidate Pool؛</li>
<li>تعداد واقعی Candidateها؛</li>
<li>نیاز یا عدم نیاز به Human Review؛</li>
<li>Identity Policy مناسب همان Unit؛</li>
<li>عدم وجود Candidate تکراری در همان Stratum.</li>
</ul>

<p>
Duplicate داخل یک Stratum مجاز نیست، اما حضور یک Source در چند Stratum مختلف ذاتاً خطا محسوب نمی‌شود.
برای مثال یک Website Article می‌تواند هم‌زمان «Direct Author Short» و Candidate یک Review Stratum نیز باشد.
بنابراین هیچ Cross-stratum Deduplication انجام نمی‌شود.
</p>

<p>
Strataهای Semantic یا Citation/QA-sensitive همچنان فقط Review Queue محسوب می‌شوند
و Candidate بودن در آن‌ها به معنی تأیید عضویت نهایی نیست.
</p>

<p>
در پایان این مرحله باید هر ۲۷ Stratum در یکی از دو وضعیت اجرایی قرار داشته باشند:
<code>AUTO_READY</code> یا <code>REVIEW_QUEUE_READY</code>.
هیچ <code>DEPENDENCY_REQUIRED</code>، Zero-match یا Unclassified Stratum نباید باقی مانده باشد.
</p>

<p>
هنوز Sampling، Question Generation، Expected Answer Generation یا Artifact Write انجام نمی‌شود.
</p>

</div>

```python
# Code Cell 19

# ============================================================
# Consolidated Golden Candidate Universe & Coverage Audit
#
# No sampling.
# No question generation.
# No artifact writes.
# ============================================================


# ============================================================
# 1) Stable identity policy for each candidate unit
# ============================================================

def golden_candidate_identity(
    source_type,
    stratum,
    item,
):

    # --------------------------------------------------------
    # WEBSITE — document-level identity
    # --------------------------------------------------------

    if source_type == "website":

        identity = {
            "document_id":
                item.get(
                    "document_id"
                ),
        }

    # --------------------------------------------------------
    # BOOK
    # --------------------------------------------------------

    elif source_type == "book":

        if stratum in {
            "intrinsically_tiny_narrative_region",
            "intrinsically_tiny_table_region",
        }:

            identity = {
                "source_id":
                    item.get(
                        "source_id"
                    ),

                "work_id":
                    item.get(
                        "work_id"
                    ),

                "edition_id":
                    item.get(
                        "edition_id"
                    ),

                "heading_path":
                    item.get(
                        "heading_path"
                    ),

                "segment_index":
                    item.get(
                        "segment_index"
                    ),

                "region_type":
                    item.get(
                        "region_type"
                    ),

                "start_paragraph":
                    item.get(
                        "start_paragraph"
                    ),

                "end_paragraph":
                    item.get(
                        "end_paragraph"
                    ),
            }

        elif stratum == "long_section":

            identity = {
                "source_id":
                    item.get(
                        "source_id"
                    ),

                "work_id":
                    item.get(
                        "work_id"
                    ),

                "edition_id":
                    item.get(
                        "edition_id"
                    ),

                "heading_path":
                    item.get(
                        "heading_path"
                    ),

                "start_paragraph":
                    item.get(
                        "start_paragraph"
                    ),

                "end_paragraph":
                    item.get(
                        "end_paragraph"
                    ),
            }

        elif stratum == "large_table_run":

            identity = {
                "work_id":
                    item.get(
                        "work_id"
                    ),

                "edition_id":
                    item.get(
                        "edition_id"
                    ),

                "start_paragraph":
                    item.get(
                        "start_paragraph"
                    ),

                "end_paragraph":
                    item.get(
                        "end_paragraph"
                    ),
            }

        elif stratum in {
            "exercise_or_self_assessment",
            "deep_heading_path",
        }:

            identity = {
                "document_id":
                    item.get(
                        "document_id"
                    ),
            }

        elif stratum == "supplemental_footnote_case":

            identity = {
                "document_id":
                    item.get(
                        "document_id"
                    ),

                "content_id":
                    item.get(
                        "content_id"
                    ),

                "parent_paragraph_id":
                    item.get(
                        "parent_paragraph_id"
                    ),
            }

        elif stratum == "multi_paragraph_evidence_case":

            identity = {
                "work_id":
                    item.get(
                        "work_id"
                    ),

                "edition_id":
                    item.get(
                        "edition_id"
                    ),

                "heading_path_key":
                    item.get(
                        "heading_path_key"
                    ),
            }

        else:

            raise AssertionError(
                "Unhandled Book stratum identity: "
                f"{stratum}"
            )

    # --------------------------------------------------------
    # TRANSCRIPT
    # --------------------------------------------------------

    elif source_type == "transcript":

        if stratum in {
            "single_speaker_primary",
            "single_speaker_unresolved",
            "multi_speaker_uni_tehran",
            "multi_speaker_chehelston",
            "multi_speaker_asre_shirin",
            "multi_speaker_khane_to",
            "multi_speaker_mizan",
            "mixed_primary_signal_case",
        }:

            identity = {
                "physical_file_id":
                    item.get(
                        "physical_file_id"
                    ),
            }

        elif stratum in {
            "primary_to_nonprimary_transition",
            "question_answer_transition",
        }:

            identity = {
                "physical_file_id":
                    item.get(
                        "physical_file_id"
                    ),

                "left_document_id":
                    item.get(
                        "left_document_id"
                    ),

                "right_document_id":
                    item.get(
                        "right_document_id"
                    ),
            }

        elif stratum in {
            "long_utterance_over_500",
            "long_utterance_over_800",
        }:

            identity = {
                "document_id":
                    item.get(
                        "document_id"
                    ),
            }

        elif stratum == "token_fallback_sentence_case":

            identity = {
                "document_id":
                    item.get(
                        "document_id"
                    ),

                "sentence_index":
                    item.get(
                        "sentence_index"
                    ),

                "char_start":
                    item.get(
                        "char_start"
                    ),

                "char_end":
                    item.get(
                        "char_end"
                    ),
            }

        else:

            raise AssertionError(
                "Unhandled Transcript stratum identity: "
                f"{stratum}"
            )

    else:

        raise AssertionError(
            f"Unknown source_type: {source_type}"
        )

    return identity


# ============================================================
# 2) Identity completeness + within-pool duplicate audit
# ============================================================

candidate_identity_audit = {}

duplicate_identity_failures = []

incomplete_identity_failures = []


for source_type in [
    "website",
    "book",
    "transcript",
]:

    candidate_identity_audit[
        source_type
    ] = {}

    for stratum in (
        expected_strata[
            source_type
        ]
    ):

        pool = candidate_pools[
            source_type
        ][
            stratum
        ]

        identity_counter = Counter()

        incomplete_count = 0

        for item in pool:

            identity = (
                golden_candidate_identity(
                    source_type,
                    stratum,
                    item,
                )
            )

            identity_complete = all(
                value is not None

                for value
                in identity.values()
            )

            if not identity_complete:

                incomplete_count += 1

                if (
                    len(
                        incomplete_identity_failures
                    )
                    < 30
                ):

                    incomplete_identity_failures.append(
                        {
                            "source_type":
                                source_type,

                            "stratum":
                                stratum,

                            "identity":
                                identity,
                        }
                    )

            identity_key = (
                stable_json_key(
                    identity
                )
            )

            identity_counter[
                identity_key
            ] += 1

        duplicate_keys = [
            identity_key
            for identity_key, count
            in identity_counter.items()
            if count > 1
        ]

        if duplicate_keys:

            duplicate_identity_failures.append(
                {
                    "source_type":
                        source_type,

                    "stratum":
                        stratum,

                    "duplicate_identity_count":
                        len(
                            duplicate_keys
                        ),

                    "examples":
                        duplicate_keys[
                            :5
                        ],
                }
            )

        candidate_identity_audit[
            source_type
        ][
            stratum
        ] = {
            "candidate_count":
                len(
                    pool
                ),

            "unique_identity_count":
                len(
                    identity_counter
                ),

            "incomplete_identity_count":
                incomplete_count,

            "duplicate_identity_count":
                len(
                    duplicate_keys
                ),
        }


# ============================================================
# 3) Consolidate all 27 strata
# ============================================================

candidate_universe_rows = []

candidate_universe_status_counts = Counter()

review_required_strata = []

auto_ready_strata = []


for source_type in [
    "website",
    "book",
    "transcript",
]:

    for stratum in (
        expected_strata[
            source_type
        ]
    ):

        rule = (
            STRATUM_RULES_V1_DRAFT[
                source_type
            ][
                stratum
            ]
        )

        state = (
            candidate_states[
                source_type
            ][
                stratum
            ]
        )

        pool_count = len(
            candidate_pools[
                source_type
            ][
                stratum
            ]
        )

        state_count = (
            state.get(
                "candidate_count"
            )
        )

        eligibility = (
            rule.get(
                "eligibility"
            )
        )

        unit = (
            rule.get(
                "unit"
            )
        )

        if (
            state[
                "status"
            ]
            == "EXACT_POOL_READY"
        ):

            execution_status = (
                "AUTO_READY"
            )

            final_membership_confirmed = (
                True
            )

            auto_ready_strata.append(
                (
                    source_type,
                    stratum,
                )
            )

        elif (
            state[
                "status"
            ]
            == "REVIEW_PREFILTER_READY"
        ):

            execution_status = (
                "REVIEW_QUEUE_READY"
            )

            final_membership_confirmed = (
                False
            )

            review_required_strata.append(
                (
                    source_type,
                    stratum,
                )
            )

        else:

            execution_status = (
                "NOT_READY"
            )

            final_membership_confirmed = (
                False
            )

        candidate_universe_status_counts[
            execution_status
        ] += 1

        identity_audit = (
            candidate_identity_audit[
                source_type
            ][
                stratum
            ]
        )

        candidate_universe_rows.append(
            {
                "source_type":
                    source_type,

                "stratum":
                    stratum,

                "unit":
                    unit,

                "eligibility":
                    eligibility,

                "candidate_state":
                    state[
                        "status"
                    ],

                "execution_status":
                    execution_status,

                "candidate_count":
                    pool_count,

                "state_candidate_count":
                    state_count,

                "unique_identity_count":
                    identity_audit[
                        "unique_identity_count"
                    ],

                "duplicate_identity_count":
                    identity_audit[
                        "duplicate_identity_count"
                    ],

                "incomplete_identity_count":
                    identity_audit[
                        "incomplete_identity_count"
                    ],

                "final_membership_confirmed":
                    final_membership_confirmed,

                "contract_rule_state":
                    GOLDEN_DATASET_CONTRACT_V1_DRAFT[
                        "stratum_rule_state"
                    ][
                        source_type
                    ][
                        stratum
                    ],
            }
        )


# ============================================================
# 4) Update in-memory rule-state registry consistently
# ============================================================

for row in (
    candidate_universe_rows
):

    source_type = (
        row[
            "source_type"
        ]
    )

    stratum = (
        row[
            "stratum"
        ]
    )

    if (
        row[
            "execution_status"
        ]
        == "AUTO_READY"
    ):

        GOLDEN_DATASET_CONTRACT_V1_DRAFT[
            "stratum_rule_state"
        ][
            source_type
        ][
            stratum
        ] = (
            "OPERATIONAL_RULE_VALIDATED"
        )

    elif (
        row[
            "execution_status"
        ]
        == "REVIEW_QUEUE_READY"
    ):

        GOLDEN_DATASET_CONTRACT_V1_DRAFT[
            "stratum_rule_state"
        ][
            source_type
        ][
            stratum
        ] = (
            "REVIEW_PREFILTER_VALIDATED_"
            "HUMAN_CONFIRMATION_REQUIRED"
        )


# refresh displayed states after update

for row in (
    candidate_universe_rows
):

    row[
        "contract_rule_state"
    ] = (
        GOLDEN_DATASET_CONTRACT_V1_DRAFT[
            "stratum_rule_state"
        ][
            row[
                "source_type"
            ]
        ][
            row[
                "stratum"
            ]
        ]
    )


# ============================================================
# 5) Candidate Universe Draft
# ============================================================

CANDIDATE_UNIVERSE_V1_DRAFT = {
    "version":
        "golden_candidate_universe_v1_draft",

    "status":
        "VALIDATED_IN_MEMORY_NOT_FROZEN",

    "contract_version":
        GOLDEN_DATASET_CONTRACT_V1_DRAFT[
            "contract_version"
        ],

    "source_of_truth": {
        "accepted_corpus_records":
            corpus_rows,

        "accepted_corpus_sha256":
            corpus_sha256,

        "measurement_tokenizer":
            measurement_model_name,

        "measurement_tokenizer_snapshot":
            resolved_tokenizer_snapshot_id,
    },

    "stratum_count":
        len(
            candidate_universe_rows
        ),

    "execution_status_counts":
        dict(
            candidate_universe_status_counts
        ),

    "auto_ready_strata":
        [
            {
                "source_type":
                    source_type,

                "stratum":
                    stratum,
            }
            for source_type, stratum
            in auto_ready_strata
        ],

    "review_required_strata":
        [
            {
                "source_type":
                    source_type,

                "stratum":
                    stratum,
            }
            for source_type, stratum
            in review_required_strata
        ],

    "strata":
        candidate_universe_rows,

    "sampling_status":
        "NOT_STARTED",

    "question_generation_status":
        "NOT_STARTED",

    "artifact_write_status":
        "NONE",
}


# ============================================================
# 6) Validation gates
# ============================================================

exact_27_strata_gate = (
    len(
        candidate_universe_rows
    )
    == 27
)


all_pools_nonempty_gate = all(
    row[
        "candidate_count"
    ]
    > 0

    for row
    in candidate_universe_rows
)


state_count_matches_pool_gate = all(
    row[
        "candidate_count"
    ]
    ==
    row[
        "state_candidate_count"
    ]

    for row
    in candidate_universe_rows
)


all_candidate_identities_complete_gate = (
    len(
        incomplete_identity_failures
    )
    == 0
)


no_within_pool_duplicate_identity_gate = (
    len(
        duplicate_identity_failures
    )
    == 0
)


all_strata_execution_ready_gate = all(
    row[
        "execution_status"
    ]
    in {
        "AUTO_READY",
        "REVIEW_QUEUE_READY",
    }

    for row
    in candidate_universe_rows
)


no_dependency_required_remains_gate = all(
    row[
        "candidate_state"
    ]
    != "DEPENDENCY_REQUIRED"

    for row
    in candidate_universe_rows
)


review_not_auto_confirmed_gate = all(
    row[
        "final_membership_confirmed"
    ]
    is False

    for row
    in candidate_universe_rows

    if (
        row[
            "execution_status"
        ]
        == "REVIEW_QUEUE_READY"
    )
)


auto_ready_confirmed_gate = all(
    row[
        "final_membership_confirmed"
    ]
    is True

    for row
    in candidate_universe_rows

    if (
        row[
            "execution_status"
        ]
        == "AUTO_READY"
    )
)


all_rule_states_resolved_gate = all(
    (
        row[
            "contract_rule_state"
        ]
        == "OPERATIONAL_RULE_VALIDATED"
    )

    if (
        row[
            "execution_status"
        ]
        == "AUTO_READY"
    )

    else
    (
        row[
            "contract_rule_state"
        ]
        ==
        "REVIEW_PREFILTER_VALIDATED_"
        "HUMAN_CONFIRMATION_REQUIRED"
    )

    for row
    in candidate_universe_rows
)


sampling_not_started_gate = (
    GOLDEN_DATASET_CONTRACT_V1_DRAFT[
        "sampling_policy"
    ][
        "status"
    ]
    == "NOT_YET_DEFINED"

    and

    CANDIDATE_UNIVERSE_V1_DRAFT[
        "sampling_status"
    ]
    == "NOT_STARTED"
)


question_generation_not_started_gate = (
    GOLDEN_DATASET_CONTRACT_V1_DRAFT[
        "question_answer_policy"
    ][
        "question_generation_status"
    ]
    == "NOT_STARTED"

    and

    CANDIDATE_UNIVERSE_V1_DRAFT[
        "question_generation_status"
    ]
    == "NOT_STARTED"
)


no_artifact_write_gate = (
    CANDIDATE_UNIVERSE_V1_DRAFT[
        "artifact_write_status"
    ]
    == "NONE"
)


cell_19_gates = {
    "exact_27_strata_consolidated":
        exact_27_strata_gate,

    "all_27_candidate_pools_nonempty":
        all_pools_nonempty_gate,

    "candidate_state_counts_match_actual_pools":
        state_count_matches_pool_gate,

    "all_candidate_identities_complete":
        all_candidate_identities_complete_gate,

    "no_duplicate_identity_within_any_pool":
        no_within_pool_duplicate_identity_gate,

    "every_stratum_is_execution_ready":
        all_strata_execution_ready_gate,

    "no_dependency_required_remains":
        no_dependency_required_remains_gate,

    "review_prefilters_not_auto_confirmed":
        review_not_auto_confirmed_gate,

    "auto_ready_memberships_confirmed":
        auto_ready_confirmed_gate,

    "all_contract_rule_states_resolved":
        all_rule_states_resolved_gate,

    "sampling_not_started":
        sampling_not_started_gate,

    "question_generation_not_started":
        question_generation_not_started_gate,

    "no_artifact_write":
        no_artifact_write_gate,
}


cell_19_pass = all(
    cell_19_gates.values()
)


# ============================================================
# 7) Output
# ============================================================

print(
    "=== CONSOLIDATED GOLDEN CANDIDATE UNIVERSE ==="
)


for source_type in [
    "website",
    "book",
    "transcript",
]:

    print(
        f"\n[{source_type.upper()}]"
    )

    for row in (
        candidate_universe_rows
    ):

        if (
            row[
                "source_type"
            ]
            != source_type
        ):
            continue

        print(
            "-",
            row[
                "stratum"
            ],
            "| unit:",
            row[
                "unit"
            ],
            "| candidates:",
            row[
                "candidate_count"
            ],
            "| status:",
            row[
                "execution_status"
            ],
            "| rule:",
            row[
                "contract_rule_state"
            ],
        )


print(
    "\n=== EXECUTION STATUS COUNTS ==="
)

for status, count in sorted(
    candidate_universe_status_counts.items()
):

    print(
        "-",
        status,
        ":",
        count,
    )


print(
    "\n=== REVIEW-REQUIRED STRATA ==="
)

for source_type, stratum in (
    review_required_strata
):

    print(
        "-",
        source_type,
        "/",
        stratum,
        "| candidates:",
        len(
            candidate_pools[
                source_type
            ][
                stratum
            ]
        ),
    )


print(
    "\n=== WITHIN-POOL IDENTITY AUDIT ==="
)

print(
    "Incomplete identity failures:",
    len(
        incomplete_identity_failures
    )
)

print(
    "Duplicate identity failures:",
    len(
        duplicate_identity_failures
    )
)


if incomplete_identity_failures:

    print(
        "\nRepresentative incomplete identities:"
    )

    for item in (
        incomplete_identity_failures[
            :10
        ]
    ):

        print(
            json.dumps(
                item,
                ensure_ascii=False,
                sort_keys=True,
            )
        )


if duplicate_identity_failures:

    print(
        "\nDuplicate identity failures:"
    )

    for item in (
        duplicate_identity_failures
    ):

        print(
            json.dumps(
                item,
                ensure_ascii=False,
                sort_keys=True,
            )
        )


print(
    "\n=== CANDIDATE UNIVERSE SUMMARY ==="
)

print(
    "Total strata:",
    CANDIDATE_UNIVERSE_V1_DRAFT[
        "stratum_count"
    ]
)

print(
    "Auto-ready strata:",
    len(
        auto_ready_strata
    )
)

print(
    "Review-required strata:",
    len(
        review_required_strata
    )
)

print(
    "Sampling status:",
    CANDIDATE_UNIVERSE_V1_DRAFT[
        "sampling_status"
    ]
)

print(
    "Question generation status:",
    CANDIDATE_UNIVERSE_V1_DRAFT[
        "question_generation_status"
    ]
)


print(
    "\n=== CODE CELL 19 GATES ==="
)

for name, passed in (
    cell_19_gates.items()
):

    print(
        name,
        ":",
        "PASS"
        if passed
        else "FAIL"
    )


print(
    "\nCODE CELL 19 GATE:",
    "PASS"
    if cell_19_pass
    else "FAIL"
)


assert cell_19_pass, (
    "Candidate Universe consolidation failed. "
    "Do not define Golden sampling policy until all 27 strata "
    "have valid, unique, recoverable candidate identities."
)
```

<div style="direction: rtl; white-space: normal; line-height: 1.7;">

<h3>Sampling Design Evidence Audit</h3>

<p>
Candidate Universe اکنون برای هر ۲۷ Stratum معتبر و قابل استفاده است.
اما تعیین تعداد Golden Caseها نباید بر اساس یک عدد دلخواه یا Threshold حدسی انجام شود.
</p>

<p>
قبل از تعریف Sampling Policy، ساختار واقعی Candidate Universe اندازه‌گیری می‌شود تا مشخص شود:
</p>

<ul>
<li>Population هر Stratum نسبت به سایر Strata چقدر کوچک یا بزرگ است؛</li>
<li>کدام Strata به‌طور طبیعی Census-like هستند؛</li>
<li>کدام Strata Population بسیار بزرگ دارند و نیازمند Selection هستند؛</li>
<li>کاندیداها در Program، Work، Edition یا Source خاص متمرکز هستند یا پراکنده؛</li>
<li>بین Strataهای یک Source چه میزان هم‌پوشانی Entity وجود دارد؛</li>
<li>چه Candidateهایی می‌توانند چند هدف ارزیابی را هم‌زمان پوشش دهند؛</li>
<li>Review Queueها چه اندازه و چه ساختاری دارند.</li>
</ul>

<p>
Overlap به‌تنهایی باعث حذف Candidate نمی‌شود.
هدف فقط این است که Sampling بعدی بتواند Coverage را افزایش و Redundancy غیرضروری را کاهش دهد.
</p>

<p>
در این Cell هیچ Quota، Sample Size، Random Seed یا Selection Rule نهایی تعریف نمی‌شود.
Sampling همچنان <code>NOT_STARTED</code> باقی می‌ماند.
</p>

</div>

```python
# Code Cell 20

# ============================================================
# Sampling Design Evidence Audit
#
# Descriptive only.
# No quota decision.
# No candidate selection.
# No random sampling.
# No question generation.
# No artifact writes.
# ============================================================


# ============================================================
# 1) Candidate-population inventory
# ============================================================

sampling_population_rows = []


for row in (
    candidate_universe_rows
):

    sampling_population_rows.append(
        {
            "source_type":
                row[
                    "source_type"
                ],

            "stratum":
                row[
                    "stratum"
                ],

            "execution_status":
                row[
                    "execution_status"
                ],

            "candidate_count":
                row[
                    "candidate_count"
                ],
        }
    )


population_counts = sorted(
    row[
        "candidate_count"
    ]
    for row
    in sampling_population_rows
)


# ============================================================
# 2) Empirical distribution helper
#
# Descriptive only — these values do NOT become thresholds.
# ============================================================

def empirical_nearest_rank(
    values,
    percentile,
):

    assert values

    ordered = sorted(
        values
    )

    rank = int(
        (
            percentile
            * len(
                ordered
            )
        )
        + 0.999999999
    )

    rank = max(
        1,
        min(
            rank,
            len(
                ordered
            ),
        ),
    )

    return ordered[
        rank - 1
    ]


population_distribution = {
    "min":
        min(
            population_counts
        ),

    "p25":
        empirical_nearest_rank(
            population_counts,
            0.25,
        ),

    "median":
        empirical_nearest_rank(
            population_counts,
            0.50,
        ),

    "p75":
        empirical_nearest_rank(
            population_counts,
            0.75,
        ),

    "p90":
        empirical_nearest_rank(
            population_counts,
            0.90,
        ),

    "max":
        max(
            population_counts
        ),
}


# ============================================================
# 3) Source-level population totals
#
# These are memberships, not unique source records.
# ============================================================

source_membership_totals = Counter()

source_stratum_counts = Counter()


for row in (
    sampling_population_rows
):

    source_type = (
        row[
            "source_type"
        ]
    )

    source_membership_totals[
        source_type
    ] += row[
        "candidate_count"
    ]

    source_stratum_counts[
        source_type
    ] += 1


# ============================================================
# 4) Generic candidate anchor extraction
#
# Anchor identities are ONLY for overlap analysis.
# They do not replace stable Golden candidate identities.
# ============================================================

def candidate_anchor_ids(
    source_type,
    stratum,
    item,
):

    anchors = set()

    # --------------------------------------------------------
    # Direct document identities
    # --------------------------------------------------------

    for field in [
        "document_id",
        "left_document_id",
        "right_document_id",
        "content_id",
        "parent_paragraph_id",
    ]:

        value = item.get(
            field
        )

        if value is not None:

            anchors.add(
                (
                    "DOCUMENT",
                    str(
                        value
                    ),
                )
            )

    # --------------------------------------------------------
    # Collections of document identities
    # --------------------------------------------------------

    for field in [
        "document_ids",
        "member_document_ids",
        "neighbor_document_ids",
    ]:

        values = (
            item.get(
                field
            )
            or []
        )

        if isinstance(
            values,
            list,
        ):

            for value in values:

                if value is not None:

                    anchors.add(
                        (
                            "DOCUMENT",
                            str(
                                value
                            ),
                        )
                    )

    # --------------------------------------------------------
    # Transcript physical-file identity
    # --------------------------------------------------------

    physical_file_id = (
        item.get(
            "physical_file_id"
        )
    )

    if (
        physical_file_id
        is not None
    ):

        anchors.add(
            (
                "PHYSICAL_FILE",
                str(
                    physical_file_id
                ),
            )
        )

    # --------------------------------------------------------
    # Book section / region identity when document IDs are not
    # the only useful structural anchor.
    # --------------------------------------------------------

    work_id = (
        item.get(
            "work_id"
        )
    )

    edition_id = (
        item.get(
            "edition_id"
        )
    )

    heading_path = (
        item.get(
            "heading_path"
        )
    )

    if (
        work_id is not None
        and
        edition_id is not None
        and
        heading_path is not None
    ):

        anchors.add(
            (
                "BOOK_SECTION",
                stable_json_key(
                    {
                        "work_id":
                            work_id,

                        "edition_id":
                            edition_id,

                        "heading_path":
                            heading_path,
                    }
                ),
            )
        )

    return anchors


# ============================================================
# 5) Build anchor sets for every stratum
# ============================================================

stratum_anchor_sets = {}


for source_type in [
    "website",
    "book",
    "transcript",
]:

    for stratum in (
        expected_strata[
            source_type
        ]
    ):

        anchors = set()

        for item in (
            candidate_pools[
                source_type
            ][
                stratum
            ]
        ):

            anchors.update(
                candidate_anchor_ids(
                    source_type,
                    stratum,
                    item,
                )
            )

        stratum_anchor_sets[
            (
                source_type,
                stratum,
            )
        ] = anchors


# ============================================================
# 6) Pairwise overlap inside each source type
# ============================================================

pairwise_overlap_rows = []


for source_type in [
    "website",
    "book",
    "transcript",
]:

    strata = list(
        expected_strata[
            source_type
        ]
    )

    for left_index in range(
        len(
            strata
        )
    ):

        for right_index in range(
            left_index + 1,
            len(
                strata
            ),
        ):

            left_stratum = (
                strata[
                    left_index
                ]
            )

            right_stratum = (
                strata[
                    right_index
                ]
            )

            left_anchors = (
                stratum_anchor_sets[
                    (
                        source_type,
                        left_stratum,
                    )
                ]
            )

            right_anchors = (
                stratum_anchor_sets[
                    (
                        source_type,
                        right_stratum,
                    )
                ]
            )

            intersection = (
                left_anchors
                &
                right_anchors
            )

            union = (
                left_anchors
                |
                right_anchors
            )

            overlap_count = len(
                intersection
            )

            if not union:

                jaccard = 0.0

            else:

                jaccard = (
                    overlap_count
                    /
                    len(
                        union
                    )
                )

            pairwise_overlap_rows.append(
                {
                    "source_type":
                        source_type,

                    "left_stratum":
                        left_stratum,

                    "right_stratum":
                        right_stratum,

                    "left_anchor_count":
                        len(
                            left_anchors
                        ),

                    "right_anchor_count":
                        len(
                            right_anchors
                        ),

                    "shared_anchor_count":
                        overlap_count,

                    "anchor_jaccard":
                        jaccard,
                }
            )


positive_overlap_rows = [
    row
    for row
    in pairwise_overlap_rows
    if (
        row[
            "shared_anchor_count"
        ]
        > 0
    )
]


positive_overlap_rows_sorted = sorted(
    positive_overlap_rows,
    key=lambda row: (
        -row[
            "shared_anchor_count"
        ],
        -row[
            "anchor_jaccard"
        ],
        row[
            "source_type"
        ],
        row[
            "left_stratum"
        ],
        row[
            "right_stratum"
        ],
    ),
)


# ============================================================
# 7) Candidate concentration by naturally available fields
# ============================================================

concentration_fields = [
    "program_id",
    "work_id",
    "edition_id",
    "source_id",
    "right_effective_label",
    "region_type",
]


stratum_concentration = {}


for source_type in [
    "website",
    "book",
    "transcript",
]:

    for stratum in (
        expected_strata[
            source_type
        ]
    ):

        pool = (
            candidate_pools[
                source_type
            ][
                stratum
            ]
        )

        field_summaries = {}

        for field in (
            concentration_fields
        ):

            counts = Counter()

            observed = 0

            for item in pool:

                value = (
                    item.get(
                        field
                    )
                )

                if value is None:
                    continue

                observed += 1

                counts[
                    scalarize(
                        value
                    )
                ] += 1

            if observed == 0:
                continue

            top_value = (
                counts.most_common(
                    1
                )[0]
            )

            field_summaries[
                field
            ] = {
                "observed_candidates":
                    observed,

                "unique_values":
                    len(
                        counts
                    ),

                "top_value":
                    top_value[
                        0
                    ],

                "top_value_count":
                    top_value[
                        1
                    ],

                "top_value_share":
                    (
                        top_value[
                            1
                        ]
                        /
                        observed
                    ),

                "top_values":
                    counts.most_common(
                        5
                    ),
            }

        stratum_concentration[
            (
                source_type,
                stratum,
            )
        ] = field_summaries


# ============================================================
# 8) Review-queue inventory
# ============================================================

review_queue_inventory = []


for source_type, stratum in (
    review_required_strata
):

    pool = (
        candidate_pools[
            source_type
        ][
            stratum
        ]
    )

    review_queue_inventory.append(
        {
            "source_type":
                source_type,

            "stratum":
                stratum,

            "candidate_count":
                len(
                    pool
                ),

            "anchor_count":
                len(
                    stratum_anchor_sets[
                        (
                            source_type,
                            stratum,
                        )
                    ]
                ),

            "final_membership_confirmed":
                False,
        }
    )


# ============================================================
# 9) Smallest / largest observed populations
#
# Descriptive ranking only.
# No cutoff is created.
# ============================================================

population_rank_ascending = sorted(
    sampling_population_rows,
    key=lambda row: (
        row[
            "candidate_count"
        ],
        row[
            "source_type"
        ],
        row[
            "stratum"
        ],
    ),
)


population_rank_descending = sorted(
    sampling_population_rows,
    key=lambda row: (
        -row[
            "candidate_count"
        ],
        row[
            "source_type"
        ],
        row[
            "stratum"
        ],
    ),
)


# ============================================================
# 10) Sampling-design evidence package
# ============================================================

SAMPLING_DESIGN_EVIDENCE_V1_DRAFT = {
    "version":
        "golden_sampling_design_evidence_v1_draft",

    "status":
        "DESCRIPTIVE_EVIDENCE_COMPLETE_NOT_POLICY",

    "stratum_count":
        len(
            sampling_population_rows
        ),

    "population_distribution":
        population_distribution,

    "source_membership_totals":
        dict(
            source_membership_totals
        ),

    "source_stratum_counts":
        dict(
            source_stratum_counts
        ),

    "review_queue_inventory":
        review_queue_inventory,

    "positive_overlap_pair_count":
        len(
            positive_overlap_rows
        ),

    "sampling_policy_status":
        "NOT_DEFINED",

    "selection_status":
        "NOT_STARTED",

    "random_seed":
        None,

    "question_generation_status":
        "NOT_STARTED",

    "artifact_write_status":
        "NONE",
}


# ============================================================
# 11) Validation gates
# ============================================================

all_27_measured_gate = (
    len(
        sampling_population_rows
    )
    == 27
)


all_population_counts_positive_gate = all(
    row[
        "candidate_count"
    ]
    > 0

    for row
    in sampling_population_rows
)


distribution_consistent_gate = (
    population_distribution[
        "min"
    ]
    ==
    min(
        population_counts
    )

    and

    population_distribution[
        "max"
    ]
    ==
    max(
        population_counts
    )

    and

    population_distribution[
        "min"
    ]
    <=
    population_distribution[
        "p25"
    ]
    <=
    population_distribution[
        "median"
    ]
    <=
    population_distribution[
        "p75"
    ]
    <=
    population_distribution[
        "p90"
    ]
    <=
    population_distribution[
        "max"
    ]
)


all_strata_have_anchor_evidence_gate = all(
    len(
        stratum_anchor_sets[
            (
                row[
                    "source_type"
                ],
                row[
                    "stratum"
                ],
            )
        ]
    )
    > 0

    for row
    in sampling_population_rows
)


review_inventory_exact_gate = (
    len(
        review_queue_inventory
    )
    ==
    len(
        review_required_strata
    )
    ==
    4
)


review_membership_not_confirmed_gate = all(
    item[
        "final_membership_confirmed"
    ]
    is False

    for item
    in review_queue_inventory
)


sampling_policy_still_undefined_gate = (
    GOLDEN_DATASET_CONTRACT_V1_DRAFT[
        "sampling_policy"
    ][
        "status"
    ]
    == "NOT_YET_DEFINED"

    and

    SAMPLING_DESIGN_EVIDENCE_V1_DRAFT[
        "sampling_policy_status"
    ]
    == "NOT_DEFINED"
)


selection_not_started_gate = (
    SAMPLING_DESIGN_EVIDENCE_V1_DRAFT[
        "selection_status"
    ]
    == "NOT_STARTED"
)


no_random_seed_declared_gate = (
    SAMPLING_DESIGN_EVIDENCE_V1_DRAFT[
        "random_seed"
    ]
    is None
)


question_generation_not_started_gate = (
    GOLDEN_DATASET_CONTRACT_V1_DRAFT[
        "question_answer_policy"
    ][
        "question_generation_status"
    ]
    == "NOT_STARTED"

    and

    SAMPLING_DESIGN_EVIDENCE_V1_DRAFT[
        "question_generation_status"
    ]
    == "NOT_STARTED"
)


no_artifact_write_gate = (
    SAMPLING_DESIGN_EVIDENCE_V1_DRAFT[
        "artifact_write_status"
    ]
    == "NONE"
)


cell_20_gates = {
    "all_27_strata_population_measured":
        all_27_measured_gate,

    "all_population_counts_positive":
        all_population_counts_positive_gate,

    "empirical_population_distribution_consistent":
        distribution_consistent_gate,

    "all_strata_have_overlap_anchor_evidence":
        all_strata_have_anchor_evidence_gate,

    "exact_4_review_queues_inventory":
        review_inventory_exact_gate,

    "review_membership_not_confirmed":
        review_membership_not_confirmed_gate,

    "sampling_policy_still_undefined":
        sampling_policy_still_undefined_gate,

    "candidate_selection_not_started":
        selection_not_started_gate,

    "random_seed_not_invented":
        no_random_seed_declared_gate,

    "question_generation_not_started":
        question_generation_not_started_gate,

    "no_artifact_write":
        no_artifact_write_gate,
}


cell_20_pass = all(
    cell_20_gates.values()
)


# ============================================================
# 12) Output
# ============================================================

print(
    "=== SAMPLING DESIGN — POPULATION DISTRIBUTION ==="
)

print(
    json.dumps(
        population_distribution,
        ensure_ascii=False,
        sort_keys=True,
    )
)


print(
    "\n=== SOURCE MEMBERSHIP TOTALS ==="
)

for source_type in [
    "website",
    "book",
    "transcript",
]:

    print(
        "-",
        source_type,
        "| strata:",
        source_stratum_counts[
            source_type
        ],
        "| candidate memberships:",
        source_membership_totals[
            source_type
        ],
    )


print(
    "\n=== SMALLEST OBSERVED STRATUM POPULATIONS ==="
)

for row in (
    population_rank_ascending[
        :10
    ]
):

    print(
        "-",
        row[
            "source_type"
        ],
        "/",
        row[
            "stratum"
        ],
        ":",
        row[
            "candidate_count"
        ],
        "|",
        row[
            "execution_status"
        ],
    )


print(
    "\n=== LARGEST OBSERVED STRATUM POPULATIONS ==="
)

for row in (
    population_rank_descending[
        :10
    ]
):

    print(
        "-",
        row[
            "source_type"
        ],
        "/",
        row[
            "stratum"
        ],
        ":",
        row[
            "candidate_count"
        ],
        "|",
        row[
            "execution_status"
        ],
    )


print(
    "\n=== REVIEW QUEUE INVENTORY ==="
)

for item in (
    review_queue_inventory
):

    print(
        json.dumps(
            item,
            ensure_ascii=False,
            sort_keys=True,
        )
    )


print(
    "\n=== TOP CROSS-STRATUM ENTITY OVERLAPS ==="
)

print(
    "Positive-overlap pairs:",
    len(
        positive_overlap_rows
    )
)

for row in (
    positive_overlap_rows_sorted[
        :20
    ]
):

    print(
        "-",
        row[
            "source_type"
        ],
        "|",
        row[
            "left_stratum"
        ],
        "<->",
        row[
            "right_stratum"
        ],
        "| shared:",
        row[
            "shared_anchor_count"
        ],
        "| jaccard:",
        round(
            row[
                "anchor_jaccard"
            ],
            4,
        ),
    )


print(
    "\n=== STRATUM CONCENTRATION SUMMARY ==="
)

for row in (
    sampling_population_rows
):

    source_type = (
        row[
            "source_type"
        ]
    )

    stratum = (
        row[
            "stratum"
        ]
    )

    summaries = (
        stratum_concentration[
            (
                source_type,
                stratum,
            )
        ]
    )

    if not summaries:
        continue

    print(
        "\n-",
        source_type,
        "/",
        stratum,
    )

    for field, summary in (
        summaries.items()
    ):

        print(
            "   ",
            field,
            "| unique:",
            summary[
                "unique_values"
            ],
            "| top:",
            summary[
                "top_value"
            ],
            "| top_count:",
            summary[
                "top_value_count"
            ],
            "| top_share:",
            round(
                summary[
                    "top_value_share"
                ],
                4,
            ),
        )


print(
    "\n=== CODE CELL 20 GATES ==="
)

for name, passed in (
    cell_20_gates.items()
):

    print(
        name,
        ":",
        "PASS"
        if passed
        else "FAIL"
    )


print(
    "\nCODE CELL 20 GATE:",
    "PASS"
    if cell_20_pass
    else "FAIL"
)


assert cell_20_pass, (
    "Sampling-design evidence audit failed. "
    "Do not define Golden quotas or select candidates until "
    "population, concentration, and overlap evidence is valid."
)
```

<div style="direction: rtl; white-space: normal; line-height: 1.7;">

<h3>Evidence-Derived Golden Sampling Policy</h3>

<p>
Population Audit نشان داد Median اندازه‌ی Strata برابر <code>38</code> Candidate است.
به‌جای انتخاب یک Sample Size دلخواه، این مقدار به‌عنوان Target مشترک Sampling استفاده می‌شود.
</p>

<p>
Policy به شکل زیر است:
</p>

<ul>
<li>اگر Population یک Auto-ready Stratum حداکثر ۳۸ باشد، کل Population به‌صورت <strong>Census</strong> نگهداری می‌شود.</li>
<li>اگر Population بزرگ‌تر از ۳۸ باشد، Target برابر ۳۸ Candidate خواهد بود.</li>
<li>Selection تصادفی نیست؛ در مرحله‌ی بعد به‌صورت <strong>Deterministic Coverage-first</strong> انجام خواهد شد.</li>
<li>هیچ Random Seed تعریف نمی‌شود.</li>
<li>برای Diversity، Fieldهای واقعی موجود مانند <code>work_id</code>، <code>edition_id</code>، <code>program_id</code> و Speaker Label فقط زمانی Constraint می‌شوند که Coverage آن‌ها در Quota ممکن باشد.</li>
<li>چهار Review Queue هنوز Final Golden Membership ندارند؛ بنابراین فقط Review Shortlist Target می‌گیرند و Final Golden Quota آن‌ها تا Human Review برابر <code>None</code> باقی می‌ماند.</li>
<li>Cross-stratum overlap باعث حذف خودکار Case نمی‌شود؛ یک Case می‌تواند چند Evaluation Label داشته باشد.</li>
</ul>

<p>
این Cell فقط Sampling Policy و Quota Plan را تعریف می‌کند.
هیچ Candidate هنوز انتخاب نمی‌شود و Question/Answer Generation همچنان شروع نمی‌شود.
</p>

</div>

```python
# Code Cell 21

# ============================================================
# Evidence-derived Sampling Policy
#
# Policy definition only.
# NO candidate selection.
# NO randomness.
# NO question generation.
# NO artifact writes.
# ============================================================


# ============================================================
# 1) Evidence-derived common target
# ============================================================

SAMPLING_TARGET = (
    population_distribution[
        "median"
    ]
)


assert SAMPLING_TARGET == 38, (
    "Sampling target must come from the measured Cell 20 "
    "population median; do not substitute a hand-picked value."
)


# ============================================================
# 2) Build per-stratum quota plan
# ============================================================

sampling_quota_plan = []


for row in (
    candidate_universe_rows
):

    source_type = (
        row[
            "source_type"
        ]
    )

    stratum = (
        row[
            "stratum"
        ]
    )

    population = (
        row[
            "candidate_count"
        ]
    )

    execution_status = (
        row[
            "execution_status"
        ]
    )

    concentration = (
        stratum_concentration.get(
            (
                source_type,
                stratum,
            ),
            {},
        )
    )

    # --------------------------------------------------------
    # AUTO-READY STRATA
    # --------------------------------------------------------

    if (
        execution_status
        == "AUTO_READY"
    ):

        if (
            population
            <= SAMPLING_TARGET
        ):

            sampling_mode = (
                "CENSUS"
            )

            planned_quota = (
                population
            )

        else:

            sampling_mode = (
                "DETERMINISTIC_COVERAGE_SELECTION"
            )

            planned_quota = (
                SAMPLING_TARGET
            )

        review_shortlist_quota = (
            None
        )

        final_golden_quota = (
            planned_quota
        )

    # --------------------------------------------------------
    # REVIEW-REQUIRED STRATA
    # --------------------------------------------------------

    elif (
        execution_status
        == "REVIEW_QUEUE_READY"
    ):

        sampling_mode = (
            "DETERMINISTIC_REVIEW_SHORTLIST"
        )

        review_shortlist_quota = min(
            population,
            SAMPLING_TARGET,
        )

        planned_quota = (
            review_shortlist_quota
        )

        # Human confirmation is still required.
        final_golden_quota = (
            None
        )

    else:

        raise AssertionError(
            "Unexpected execution status: "
            f"{execution_status}"
        )


    # ========================================================
    # Diversity dimensions derived from actual observations.
    #
    # Guarantee complete categorical coverage only when:
    # - there is more than one observed value;
    # - the number of values fits within the planned quota.
    #
    # Otherwise the field may still be used as a
    # coverage-maximization dimension, but not as a hard
    # all-values requirement.
    # ========================================================

    guaranteed_diversity_fields = []

    maximize_diversity_fields = []


    for field, summary in (
        concentration.items()
    ):

        unique_values = (
            summary[
                "unique_values"
            ]
        )

        if (
            unique_values
            <= 1
        ):
            continue

        if (
            unique_values
            <= planned_quota
        ):

            guaranteed_diversity_fields.append(
                field
            )

        else:

            maximize_diversity_fields.append(
                field
            )


    # ========================================================
    # Stable deterministic tie-break rule
    # ========================================================

    deterministic_tie_break = {
        "method":
            "SHA256_OF_STABLE_CANDIDATE_IDENTITY",

        "random_seed":
            None,

        "random_sampling":
            False,
    }


    sampling_quota_plan.append(
        {
            "source_type":
                source_type,

            "stratum":
                stratum,

            "execution_status":
                execution_status,

            "population":
                population,

            "sampling_mode":
                sampling_mode,

            "planned_selection_quota":
                planned_quota,

            "review_shortlist_quota":
                review_shortlist_quota,

            "final_golden_quota":
                final_golden_quota,

            "guaranteed_diversity_fields":
                sorted(
                    guaranteed_diversity_fields
                ),

            "maximize_diversity_fields":
                sorted(
                    maximize_diversity_fields
                ),

            "deterministic_tie_break":
                deterministic_tie_break,

            "selection_status":
                "NOT_STARTED",
        }
    )


# ============================================================
# 3) Sampling-policy summary
# ============================================================

sampling_mode_counts = Counter(
    row[
        "sampling_mode"
    ]
    for row
    in sampling_quota_plan
)


planned_auto_memberships = sum(
    row[
        "planned_selection_quota"
    ]

    for row
    in sampling_quota_plan

    if (
        row[
            "execution_status"
        ]
        == "AUTO_READY"
    )
)


planned_review_shortlist_memberships = sum(
    row[
        "review_shortlist_quota"
    ]
    or 0

    for row
    in sampling_quota_plan

    if (
        row[
            "execution_status"
        ]
        == "REVIEW_QUEUE_READY"
    )
)


census_strata = [
    row
    for row
    in sampling_quota_plan
    if (
        row[
            "sampling_mode"
        ]
        == "CENSUS"
    )
]


selection_strata = [
    row
    for row
    in sampling_quota_plan
    if (
        row[
            "sampling_mode"
        ]
        == "DETERMINISTIC_COVERAGE_SELECTION"
    )
]


review_shortlist_strata = [
    row
    for row
    in sampling_quota_plan
    if (
        row[
            "sampling_mode"
        ]
        == "DETERMINISTIC_REVIEW_SHORTLIST"
    )
]


# ============================================================
# 4) Sampling policy contract
# ============================================================

sampling_policy_existing = dict(
    GOLDEN_DATASET_CONTRACT_V1_DRAFT.get(
        "sampling_policy",
        {},
    )
)


sampling_policy_existing.update(
    {
        "status":
            "DEFINED_NOT_EXECUTED",

        "policy_version":
            "golden_sampling_policy_v1_draft",

        "target_basis":
            "EMPIRICAL_MEDIAN_STRATUM_POPULATION",

        "target_quota":
            SAMPLING_TARGET,

        "small_population_policy":
            "CENSUS_WHEN_POPULATION_LE_TARGET",

        "large_population_policy":
            "DETERMINISTIC_COVERAGE_FIRST_SELECTION",

        "review_queue_policy":
            (
                "DETERMINISTIC_SHORTLIST_FOR_HUMAN_CONFIRMATION;"
                "NO_FINAL_GOLDEN_MEMBERSHIP_BEFORE_REVIEW"
            ),

        "cross_stratum_overlap_policy":
            (
                "ALLOWED;"
                "DO_NOT_AUTO_DEDUP_ACROSS_STRATA;"
                "PRESERVE_ALL_EVALUATION_LABELS"
            ),

        "diversity_policy":
            {
                "hard_coverage":
                    (
                        "COVER_ALL_OBSERVED_VALUES_WHEN_"
                        "UNIQUE_VALUE_COUNT_FITS_WITHIN_QUOTA"
                    ),

                "otherwise":
                    "MAXIMIZE_CATEGORICAL_DIVERSITY",

                "observed_fields_only":
                    True,
            },

        "determinism":
            {
                "random_sampling":
                    False,

                "random_seed":
                    None,

                "tie_break":
                    "SHA256_OF_STABLE_CANDIDATE_IDENTITY",
            },

        "selection_status":
            "NOT_STARTED",
    }
)


GOLDEN_DATASET_CONTRACT_V1_DRAFT[
    "sampling_policy"
] = sampling_policy_existing


CANDIDATE_UNIVERSE_V1_DRAFT[
    "sampling_status"
] = (
    "POLICY_DEFINED_SELECTION_NOT_STARTED"
)


# ============================================================
# 5) Sampling policy draft object
# ============================================================

SAMPLING_POLICY_V1_DRAFT = {
    "version":
        "golden_sampling_policy_v1_draft",

    "status":
        "VALIDATED_IN_MEMORY_NOT_EXECUTED",

    "evidence_source":
        "CODE_CELL_20_SAMPLING_DESIGN_EVIDENCE",

    "population_median":
        population_distribution[
            "median"
        ],

    "target_quota":
        SAMPLING_TARGET,

    "stratum_count":
        len(
            sampling_quota_plan
        ),

    "mode_counts":
        dict(
            sampling_mode_counts
        ),

    "planned_auto_golden_memberships":
        planned_auto_memberships,

    "planned_review_shortlist_memberships":
        planned_review_shortlist_memberships,

    "quota_plan":
        sampling_quota_plan,

    "selection_status":
        "NOT_STARTED",

    "question_generation_status":
        "NOT_STARTED",

    "artifact_write_status":
        "NONE",
}


# ============================================================
# 6) Independent validation gates
# ============================================================

exact_27_quota_rows_gate = (
    len(
        sampling_quota_plan
    )
    == 27
)


target_is_empirical_median_gate = (
    SAMPLING_TARGET
    ==
    population_distribution[
        "median"
    ]
    ==
    38
)


all_planned_quotas_positive_gate = all(
    row[
        "planned_selection_quota"
    ]
    > 0

    for row
    in sampling_quota_plan
)


no_quota_exceeds_population_gate = all(
    row[
        "planned_selection_quota"
    ]
    <=
    row[
        "population"
    ]

    for row
    in sampling_quota_plan
)


auto_census_rule_gate = all(
    (
        row[
            "sampling_mode"
        ]
        == "CENSUS"

        and

        row[
            "planned_selection_quota"
        ]
        ==
        row[
            "population"
        ]
    )

    if (
        row[
            "execution_status"
        ]
        == "AUTO_READY"

        and

        row[
            "population"
        ]
        <= SAMPLING_TARGET
    )

    else True

    for row
    in sampling_quota_plan
)


auto_large_rule_gate = all(
    (
        row[
            "sampling_mode"
        ]
        ==
        "DETERMINISTIC_COVERAGE_SELECTION"

        and

        row[
            "planned_selection_quota"
        ]
        ==
        SAMPLING_TARGET
    )

    if (
        row[
            "execution_status"
        ]
        == "AUTO_READY"

        and

        row[
            "population"
        ]
        > SAMPLING_TARGET
    )

    else True

    for row
    in sampling_quota_plan
)


review_final_quota_unset_gate = all(
    (
        row[
            "review_shortlist_quota"
        ]
        ==
        min(
            row[
                "population"
            ],
            SAMPLING_TARGET,
        )

        and

        row[
            "final_golden_quota"
        ]
        is None
    )

    for row
    in sampling_quota_plan

    if (
        row[
            "execution_status"
        ]
        == "REVIEW_QUEUE_READY"
    )
)


review_count_four_gate = (
    len(
        review_shortlist_strata
    )
    == 4
)


deterministic_no_randomness_gate = all(
    row[
        "deterministic_tie_break"
    ][
        "random_sampling"
    ]
    is False

    and

    row[
        "deterministic_tie_break"
    ][
        "random_seed"
    ]
    is None

    for row
    in sampling_quota_plan
)


diversity_fields_evidence_based_gate = all(
    field
    in stratum_concentration[
        (
            row[
                "source_type"
            ],
            row[
                "stratum"
            ],
        )
    ]

    for row
    in sampling_quota_plan

    for field
    in (
        row[
            "guaranteed_diversity_fields"
        ]
        +
        row[
            "maximize_diversity_fields"
        ]
    )
)


selection_not_started_gate = all(
    row[
        "selection_status"
    ]
    == "NOT_STARTED"

    for row
    in sampling_quota_plan
)


question_generation_not_started_gate = (
    GOLDEN_DATASET_CONTRACT_V1_DRAFT[
        "question_answer_policy"
    ][
        "question_generation_status"
    ]
    == "NOT_STARTED"

    and

    SAMPLING_POLICY_V1_DRAFT[
        "question_generation_status"
    ]
    == "NOT_STARTED"
)


no_artifact_write_gate = (
    SAMPLING_POLICY_V1_DRAFT[
        "artifact_write_status"
    ]
    == "NONE"
)


cell_21_gates = {
    "exact_27_sampling_quota_rows":
        exact_27_quota_rows_gate,

    "target_quota_is_empirical_median_38":
        target_is_empirical_median_gate,

    "all_planned_quotas_positive":
        all_planned_quotas_positive_gate,

    "no_quota_exceeds_population":
        no_quota_exceeds_population_gate,

    "auto_small_population_census_rule":
        auto_census_rule_gate,

    "auto_large_population_target_38_rule":
        auto_large_rule_gate,

    "review_final_golden_quota_remains_unset":
        review_final_quota_unset_gate,

    "exact_4_review_shortlist_strata":
        review_count_four_gate,

    "deterministic_without_random_seed":
        deterministic_no_randomness_gate,

    "diversity_dimensions_derived_from_observed_fields":
        diversity_fields_evidence_based_gate,

    "candidate_selection_not_started":
        selection_not_started_gate,

    "question_generation_not_started":
        question_generation_not_started_gate,

    "no_artifact_write":
        no_artifact_write_gate,
}


cell_21_pass = all(
    cell_21_gates.values()
)


# ============================================================
# 7) Output
# ============================================================

print(
    "=== GOLDEN SAMPLING POLICY ==="
)

print(
    "Evidence-derived target quota:",
    SAMPLING_TARGET
)

print(
    "Target basis:",
    "EMPIRICAL_MEDIAN_STRATUM_POPULATION"
)

print(
    "Auto planned memberships:",
    planned_auto_memberships
)

print(
    "Review shortlist memberships:",
    planned_review_shortlist_memberships
)


print(
    "\n=== SAMPLING MODE COUNTS ==="
)

for mode, count in sorted(
    sampling_mode_counts.items()
):

    print(
        "-",
        mode,
        ":",
        count,
    )


print(
    "\n=== PER-STRATUM QUOTA PLAN ==="
)

for row in (
    sampling_quota_plan
):

    print(
        "-",
        row[
            "source_type"
        ],
        "/",
        row[
            "stratum"
        ],
        "| population:",
        row[
            "population"
        ],
        "| mode:",
        row[
            "sampling_mode"
        ],
        "| planned:",
        row[
            "planned_selection_quota"
        ],
        "| final_golden:",
        row[
            "final_golden_quota"
        ],
        "| guaranteed_diversity:",
        row[
            "guaranteed_diversity_fields"
        ],
        "| maximize_diversity:",
        row[
            "maximize_diversity_fields"
        ],
    )


print(
    "\n=== POLICY SUMMARY ==="
)

print(
    "Census strata:",
    len(
        census_strata
    )
)

print(
    "Coverage-selection strata:",
    len(
        selection_strata
    )
)

print(
    "Review-shortlist strata:",
    len(
        review_shortlist_strata
    )
)

print(
    "Selection status:",
    SAMPLING_POLICY_V1_DRAFT[
        "selection_status"
    ]
)

print(
    "Question generation:",
    SAMPLING_POLICY_V1_DRAFT[
        "question_generation_status"
    ]
)


print(
    "\n=== CODE CELL 21 GATES ==="
)

for name, passed in (
    cell_21_gates.items()
):

    print(
        name,
        ":",
        "PASS"
        if passed
        else "FAIL"
    )


print(
    "\nCODE CELL 21 GATE:",
    "PASS"
    if cell_21_pass
    else "FAIL"
)


assert cell_21_pass, (
    "Golden Sampling Policy validation failed. "
    "Do not select candidates until the evidence-derived "
    "quota policy is internally consistent."
)
```

<div style="direction: rtl; white-space: normal; line-height: 1.7;">

<h3>Execute Deterministic Coverage-First Candidate Selection</h3>

<p>
Sampling Policy اکنون تعریف شده است و برای هر ۲۷ Stratum یک Quota معتبر دارد.
در این مرحله برای اولین بار Candidate Selection واقعاً اجرا می‌شود.
</p>

<p>
Selection هیچ عنصر تصادفی ندارد.
برای هر Candidate ابتدا Stable Candidate Identity ساخته می‌شود و سپس SHA256 آن
به‌عنوان Tie-break قطعی استفاده می‌شود.
</p>

<p>
در Strataهای <code>CENSUS</code> کل Population نگهداری می‌شود.
در Strataهای بزرگ‌تر Selection به‌صورت Coverage-first انجام می‌شود:
ابتدا پوشش Diversity Fieldهای تضمین‌شده بیشینه می‌شود،
سپس Diversity Fieldهای قابل‌افزایش، و در نهایت SHA256 Identity تصمیم Tie-break را می‌گیرد.
</p>

<p>
چهار Review Queue نیز فقط Shortlist می‌شوند.
عضویت این موارد در Golden Dataset هنوز تأیید نشده و تا Human Review
نباید به <code>Final Golden Membership</code> تبدیل شود.
</p>

<p>
Cross-stratum overlap مجاز باقی می‌ماند و هیچ Case به دلیل حضور در Stratum دیگر حذف نمی‌شود.
همچنین هنوز Question Generation، Expected Answer Generation یا Artifact Write انجام نمی‌شود.
</p>

</div>

```python
# Code Cell 22

# ============================================================
# Deterministic Coverage-First Selection
#
# Executes the sampling policy from Code Cell 21.
#
# NO randomness.
# NO cross-stratum deduplication.
# NO question generation.
# NO artifact writes.
# ============================================================


# ============================================================
# 1) Stable candidate identity + deterministic hash
# ============================================================

def stable_candidate_identity_record(
    source_type,
    stratum,
    item,
):

    identity = golden_candidate_identity(
        source_type,
        stratum,
        item,
    )

    identity_json = stable_json_key(
        identity
    )

    identity_sha256 = hashlib.sha256(
        identity_json.encode(
            "utf-8"
        )
    ).hexdigest()

    return {
        "identity":
            identity,

        "identity_json":
            identity_json,

        "identity_sha256":
            identity_sha256,
    }


# ============================================================
# 2) Normalize diversity-field values
# ============================================================

def diversity_values(
    value,
):

    if value is None:
        return set()

    if isinstance(
        value,
        dict,
    ):

        return {
            stable_json_key(
                value
            )
        }

    if isinstance(
        value,
        (
            list,
            tuple,
            set,
        ),
    ):

        normalized = set()

        for member in value:

            if member is None:
                continue

            if isinstance(
                member,
                (
                    dict,
                    list,
                    tuple,
                ),
            ):

                normalized.add(
                    stable_json_key(
                        member
                    )
                )

            else:

                normalized.add(
                    str(
                        scalarize(
                            member
                        )
                    )
                )

        return normalized

    return {
        str(
            scalarize(
                value
            )
        )
    }


def observed_diversity_values(
    pool,
    field,
):

    observed = set()

    for item in pool:

        observed.update(
            diversity_values(
                item.get(
                    field
                )
            )
        )

    return observed


# ============================================================
# 3) Deterministic coverage-first selector
# ============================================================

def deterministic_coverage_select(
    source_type,
    stratum,
    pool,
    quota,
    guaranteed_fields,
    maximize_fields,
):

    assert (
        quota
        > 0
    )

    assert (
        quota
        <= len(
            pool
        )
    )

    enriched = []

    for source_index, item in enumerate(
        pool
    ):

        identity_record = (
            stable_candidate_identity_record(
                source_type,
                stratum,
                item,
            )
        )

        field_values = {}

        for field in (
            guaranteed_fields
            +
            maximize_fields
        ):

            field_values[
                field
            ] = diversity_values(
                item.get(
                    field
                )
            )

        enriched.append(
            {
                "source_index":
                    source_index,

                "candidate":
                    item,

                "identity":
                    identity_record[
                        "identity"
                    ],

                "identity_json":
                    identity_record[
                        "identity_json"
                    ],

                "identity_sha256":
                    identity_record[
                        "identity_sha256"
                    ],

                "field_values":
                    field_values,
            }
        )


    # --------------------------------------------------------
    # Census: include everything, with deterministic ranking.
    # --------------------------------------------------------

    if (
        quota
        ==
        len(
            enriched
        )
    ):

        ordered = sorted(
            enriched,
            key=lambda item:
                item[
                    "identity_sha256"
                ],
        )

        selected = []

        for rank, item in enumerate(
            ordered,
            start=1,
        ):

            selected.append(
                {
                    **item,
                    "selection_rank":
                        rank,

                    "guaranteed_gain_at_selection":
                        None,

                    "maximize_gain_at_selection":
                        None,
                }
            )

        return selected


    # --------------------------------------------------------
    # Coverage-first selection.
    # --------------------------------------------------------

    covered_guaranteed = {
        field:
            set()

        for field
        in guaranteed_fields
    }

    covered_maximize = {
        field:
            set()

        for field
        in maximize_fields
    }


    remaining = list(
        enriched
    )

    selected = []


    while (
        len(
            selected
        )
        < quota
    ):

        scored = []

        for item in (
            remaining
        ):

            guaranteed_gain = sum(
                len(
                    item[
                        "field_values"
                    ][
                        field
                    ]
                    -
                    covered_guaranteed[
                        field
                    ]
                )

                for field
                in guaranteed_fields
            )


            maximize_gain = sum(
                len(
                    item[
                        "field_values"
                    ][
                        field
                    ]
                    -
                    covered_maximize[
                        field
                    ]
                )

                for field
                in maximize_fields
            )


            scored.append(
                (
                    -guaranteed_gain,
                    -maximize_gain,
                    item[
                        "identity_sha256"
                    ],
                    guaranteed_gain,
                    maximize_gain,
                    item,
                )
            )


        scored.sort(
            key=lambda row: (
                row[
                    0
                ],
                row[
                    1
                ],
                row[
                    2
                ],
            )
        )


        (
            _,
            _,
            _,
            guaranteed_gain,
            maximize_gain,
            winner,
        ) = scored[
            0
        ]


        for field in (
            guaranteed_fields
        ):

            covered_guaranteed[
                field
            ].update(
                winner[
                    "field_values"
                ][
                    field
                ]
            )


        for field in (
            maximize_fields
        ):

            covered_maximize[
                field
            ].update(
                winner[
                    "field_values"
                ][
                    field
                ]
            )


        selected.append(
            {
                **winner,

                "selection_rank":
                    len(
                        selected
                    )
                    + 1,

                "guaranteed_gain_at_selection":
                    guaranteed_gain,

                "maximize_gain_at_selection":
                    maximize_gain,
            }
        )


        winner_sha = (
            winner[
                "identity_sha256"
            ]
        )

        remaining = [
            item
            for item
            in remaining
            if (
                item[
                    "identity_sha256"
                ]
                != winner_sha
            )
        ]


    return selected


# ============================================================
# 4) Execute policy for all 27 strata
# ============================================================

selected_candidates = {
    "website": {},
    "book": {},
    "transcript": {},
}


selection_summary_rows = []


for quota_row in (
    sampling_quota_plan
):

    source_type = (
        quota_row[
            "source_type"
        ]
    )

    stratum = (
        quota_row[
            "stratum"
        ]
    )

    pool = (
        candidate_pools[
            source_type
        ][
            stratum
        ]
    )

    quota = (
        quota_row[
            "planned_selection_quota"
        ]
    )

    guaranteed_fields = (
        quota_row[
            "guaranteed_diversity_fields"
        ]
    )

    maximize_fields = (
        quota_row[
            "maximize_diversity_fields"
        ]
    )


    selected = (
        deterministic_coverage_select(
            source_type=(
                source_type
            ),
            stratum=(
                stratum
            ),
            pool=(
                pool
            ),
            quota=(
                quota
            ),
            guaranteed_fields=(
                guaranteed_fields
            ),
            maximize_fields=(
                maximize_fields
            ),
        )
    )


    # --------------------------------------------------------
    # Final membership state
    # --------------------------------------------------------

    if (
        quota_row[
            "execution_status"
        ]
        == "AUTO_READY"
    ):

        membership_status = (
            "AUTO_GOLDEN_SELECTED"
        )

        final_golden_membership = (
            True
        )

    else:

        membership_status = (
            "REVIEW_SHORTLIST_PENDING_HUMAN_CONFIRMATION"
        )

        final_golden_membership = (
            False
        )


    selected_records = []


    for item in (
        selected
    ):

        selected_records.append(
            {
                "source_type":
                    source_type,

                "stratum":
                    stratum,

                "selection_rank":
                    item[
                        "selection_rank"
                    ],

                "candidate_identity":
                    item[
                        "identity"
                    ],

                "candidate_identity_sha256":
                    item[
                        "identity_sha256"
                    ],

                "membership_status":
                    membership_status,

                "final_golden_membership":
                    final_golden_membership,

                "candidate":
                    item[
                        "candidate"
                    ],

                "coverage_gain": {
                    "guaranteed":
                        item[
                            "guaranteed_gain_at_selection"
                        ],

                    "maximize":
                        item[
                            "maximize_gain_at_selection"
                        ],
                },
            }
        )


    selected_candidates[
        source_type
    ][
        stratum
    ] = (
        selected_records
    )


    # --------------------------------------------------------
    # Diversity coverage audit
    # --------------------------------------------------------

    guaranteed_coverage = {}

    maximize_coverage = {}


    for field in (
        guaranteed_fields
    ):

        population_values = (
            observed_diversity_values(
                pool,
                field,
            )
        )

        selected_values = set()

        for selected_item in (
            selected_records
        ):

            selected_values.update(
                diversity_values(
                    selected_item[
                        "candidate"
                    ].get(
                        field
                    )
                )
            )

        guaranteed_coverage[
            field
        ] = {
            "population_unique_values":
                len(
                    population_values
                ),

            "selected_unique_values":
                len(
                    selected_values
                ),

            "all_population_values_covered":
                (
                    selected_values
                    ==
                    population_values
                ),
        }


    for field in (
        maximize_fields
    ):

        population_values = (
            observed_diversity_values(
                pool,
                field,
            )
        )

        selected_values = set()

        for selected_item in (
            selected_records
        ):

            selected_values.update(
                diversity_values(
                    selected_item[
                        "candidate"
                    ].get(
                        field
                    )
                )
            )

        maximize_coverage[
            field
        ] = {
            "population_unique_values":
                len(
                    population_values
                ),

            "selected_unique_values":
                len(
                    selected_values
                ),

            "selected_share_of_population_values":
                (
                    len(
                        selected_values
                    )
                    /
                    len(
                        population_values
                    )

                    if population_values
                    else 1.0
                ),
        }


    selection_summary_rows.append(
        {
            "source_type":
                source_type,

            "stratum":
                stratum,

            "population":
                len(
                    pool
                ),

            "planned_quota":
                quota,

            "selected_count":
                len(
                    selected_records
                ),

            "sampling_mode":
                quota_row[
                    "sampling_mode"
                ],

            "execution_status":
                quota_row[
                    "execution_status"
                ],

            "membership_status":
                membership_status,

            "final_golden_membership":
                final_golden_membership,

            "guaranteed_diversity_coverage":
                guaranteed_coverage,

            "maximize_diversity_coverage":
                maximize_coverage,
        }
    )


# ============================================================
# 5) Independent deterministic recomputation
# ============================================================

deterministic_recheck_failures = []


for quota_row in (
    sampling_quota_plan
):

    source_type = (
        quota_row[
            "source_type"
        ]
    )

    stratum = (
        quota_row[
            "stratum"
        ]
    )

    pool = (
        candidate_pools[
            source_type
        ][
            stratum
        ]
    )


    second_pass = (
        deterministic_coverage_select(
            source_type=(
                source_type
            ),
            stratum=(
                stratum
            ),
            pool=(
                pool
            ),
            quota=(
                quota_row[
                    "planned_selection_quota"
                ]
            ),
            guaranteed_fields=(
                quota_row[
                    "guaranteed_diversity_fields"
                ]
            ),
            maximize_fields=(
                quota_row[
                    "maximize_diversity_fields"
                ]
            ),
        )
    )


    first_hashes = [
        item[
            "candidate_identity_sha256"
        ]
        for item
        in selected_candidates[
            source_type
        ][
            stratum
        ]
    ]


    second_hashes = [
        item[
            "identity_sha256"
        ]
        for item
        in second_pass
    ]


    if (
        first_hashes
        != second_hashes
    ):

        deterministic_recheck_failures.append(
            {
                "source_type":
                    source_type,

                "stratum":
                    stratum,
            }
        )


# ============================================================
# 6) Identity uniqueness audit inside selected strata
# ============================================================

selected_duplicate_identity_failures = []


for source_type in [
    "website",
    "book",
    "transcript",
]:

    for stratum in (
        expected_strata[
            source_type
        ]
    ):

        selected = (
            selected_candidates[
                source_type
            ][
                stratum
            ]
        )

        hashes = [
            item[
                "candidate_identity_sha256"
            ]
            for item
            in selected
        ]

        if (
            len(
                hashes
            )
            !=
            len(
                set(
                    hashes
                )
            )
        ):

            selected_duplicate_identity_failures.append(
                {
                    "source_type":
                        source_type,

                    "stratum":
                        stratum,
                }
            )


# ============================================================
# 7) Aggregate selection counts
# ============================================================

selected_auto_memberships = sum(
    row[
        "selected_count"
    ]

    for row
    in selection_summary_rows

    if (
        row[
            "execution_status"
        ]
        == "AUTO_READY"
    )
)


selected_review_shortlist_memberships = sum(
    row[
        "selected_count"
    ]

    for row
    in selection_summary_rows

    if (
        row[
            "execution_status"
        ]
        == "REVIEW_QUEUE_READY"
    )
)


selected_total_memberships = (
    selected_auto_memberships
    +
    selected_review_shortlist_memberships
)


# ============================================================
# 8) Golden selection draft
# ============================================================

GOLDEN_SELECTION_V1_DRAFT = {
    "version":
        "golden_selection_v1_draft",

    "status":
        "SELECTED_IN_MEMORY_NOT_FROZEN",

    "sampling_policy_version":
        SAMPLING_POLICY_V1_DRAFT[
            "version"
        ],

    "sampling_target":
        SAMPLING_TARGET,

    "selection_method":
        "DETERMINISTIC_COVERAGE_FIRST",

    "random_sampling":
        False,

    "random_seed":
        None,

    "tie_break":
        "SHA256_OF_STABLE_CANDIDATE_IDENTITY",

    "auto_golden_memberships":
        selected_auto_memberships,

    "review_shortlist_memberships":
        selected_review_shortlist_memberships,

    "total_selected_memberships":
        selected_total_memberships,

    "strata":
        selection_summary_rows,

    "question_generation_status":
        "NOT_STARTED",

    "artifact_write_status":
        "NONE",
}


# ============================================================
# 9) Validation gates
# ============================================================

exact_27_selected_strata_gate = (
    len(
        selection_summary_rows
    )
    == 27
)


all_selection_counts_match_quota_gate = all(
    row[
        "selected_count"
    ]
    ==
    row[
        "planned_quota"
    ]

    for row
    in selection_summary_rows
)


auto_total_matches_plan_gate = (
    selected_auto_memberships
    ==
    planned_auto_memberships
    ==
    572
)


review_total_matches_plan_gate = (
    selected_review_shortlist_memberships
    ==
    planned_review_shortlist_memberships
    ==
    152
)


census_exact_population_gate = all(
    row[
        "selected_count"
    ]
    ==
    row[
        "population"
    ]

    for row
    in selection_summary_rows

    if (
        row[
            "sampling_mode"
        ]
        == "CENSUS"
    )
)


guaranteed_diversity_complete_gate = all(
    coverage[
        "all_population_values_covered"
    ]

    for row
    in selection_summary_rows

    for coverage
    in row[
        "guaranteed_diversity_coverage"
    ].values()
)


deterministic_recomputation_gate = (
    len(
        deterministic_recheck_failures
    )
    == 0
)


selected_identity_unique_gate = (
    len(
        selected_duplicate_identity_failures
    )
    == 0
)


review_not_final_gate = all(
    row[
        "final_golden_membership"
    ]
    is False

    for row
    in selection_summary_rows

    if (
        row[
            "execution_status"
        ]
        == "REVIEW_QUEUE_READY"
    )
)


auto_membership_true_gate = all(
    row[
        "final_golden_membership"
    ]
    is True

    for row
    in selection_summary_rows

    if (
        row[
            "execution_status"
        ]
        == "AUTO_READY"
    )
)


no_randomness_gate = (
    GOLDEN_SELECTION_V1_DRAFT[
        "random_sampling"
    ]
    is False

    and

    GOLDEN_SELECTION_V1_DRAFT[
        "random_seed"
    ]
    is None
)


question_generation_not_started_gate = (
    GOLDEN_DATASET_CONTRACT_V1_DRAFT[
        "question_answer_policy"
    ][
        "question_generation_status"
    ]
    == "NOT_STARTED"

    and

    GOLDEN_SELECTION_V1_DRAFT[
        "question_generation_status"
    ]
    == "NOT_STARTED"
)


no_artifact_write_gate = (
    GOLDEN_SELECTION_V1_DRAFT[
        "artifact_write_status"
    ]
    == "NONE"
)


cell_22_gates = {
    "exact_27_strata_selected":
        exact_27_selected_strata_gate,

    "every_selection_matches_planned_quota":
        all_selection_counts_match_quota_gate,

    "auto_golden_memberships_match_plan_572":
        auto_total_matches_plan_gate,

    "review_shortlist_memberships_match_plan_152":
        review_total_matches_plan_gate,

    "census_strata_include_full_population":
        census_exact_population_gate,

    "all_guaranteed_diversity_values_covered":
        guaranteed_diversity_complete_gate,

    "independent_deterministic_recomputation_matches":
        deterministic_recomputation_gate,

    "no_duplicate_identity_within_selected_strata":
        selected_identity_unique_gate,

    "review_shortlists_not_final_golden":
        review_not_final_gate,

    "auto_selected_memberships_are_golden":
        auto_membership_true_gate,

    "selection_contains_no_randomness":
        no_randomness_gate,

    "question_generation_not_started":
        question_generation_not_started_gate,

    "no_artifact_write":
        no_artifact_write_gate,
}


cell_22_pass = all(
    cell_22_gates.values()
)


# ============================================================
# 10) Only after successful validation, update in-memory status
# ============================================================

if cell_22_pass:

    GOLDEN_DATASET_CONTRACT_V1_DRAFT[
        "sampling_policy"
    ][
        "selection_status"
    ] = (
        "EXECUTED_IN_MEMORY_NOT_FROZEN"
    )

    CANDIDATE_UNIVERSE_V1_DRAFT[
        "sampling_status"
    ] = (
        "SELECTION_EXECUTED_IN_MEMORY_NOT_FROZEN"
    )

    SAMPLING_POLICY_V1_DRAFT[
        "selection_status"
    ] = (
        "EXECUTED_IN_MEMORY_NOT_FROZEN"
    )


# ============================================================
# 11) Output
# ============================================================

print(
    "=== DETERMINISTIC GOLDEN SELECTION ==="
)

print(
    "Auto Golden memberships:",
    selected_auto_memberships
)

print(
    "Review shortlist memberships:",
    selected_review_shortlist_memberships
)

print(
    "Total selected memberships:",
    selected_total_memberships
)


print(
    "\n=== PER-STRATUM SELECTION ==="
)

for row in (
    selection_summary_rows
):

    print(
        "-",
        row[
            "source_type"
        ],
        "/",
        row[
            "stratum"
        ],
        "| population:",
        row[
            "population"
        ],
        "| selected:",
        row[
            "selected_count"
        ],
        "| mode:",
        row[
            "sampling_mode"
        ],
        "| membership:",
        row[
            "membership_status"
        ],
    )

    for field, coverage in (
        row[
            "guaranteed_diversity_coverage"
        ].items()
    ):

        print(
            "   guaranteed:",
            field,
            "| population_values:",
            coverage[
                "population_unique_values"
            ],
            "| selected_values:",
            coverage[
                "selected_unique_values"
            ],
            "| complete:",
            coverage[
                "all_population_values_covered"
            ],
        )

    for field, coverage in (
        row[
            "maximize_diversity_coverage"
        ].items()
    ):

        print(
            "   maximize:",
            field,
            "| population_values:",
            coverage[
                "population_unique_values"
            ],
            "| selected_values:",
            coverage[
                "selected_unique_values"
            ],
            "| share:",
            round(
                coverage[
                    "selected_share_of_population_values"
                ],
                4,
            ),
        )


print(
    "\n=== REPRESENTATIVE SELECTED IDENTITIES ==="
)

for source_type in [
    "website",
    "book",
    "transcript",
]:

    print(
        f"\n[{source_type.upper()}]"
    )

    for stratum in (
        expected_strata[
            source_type
        ]
    ):

        selected = (
            selected_candidates[
                source_type
            ][
                stratum
            ]
        )

        print(
            "-",
            stratum,
            "| selected:",
            len(
                selected
            ),
        )

        for item in (
            selected[
                :2
            ]
        ):

            print(
                "   ",
                json.dumps(
                    {
                        "selection_rank":
                            item[
                                "selection_rank"
                            ],

                        "candidate_identity":
                            item[
                                "candidate_identity"
                            ],

                        "candidate_identity_sha256":
                            item[
                                "candidate_identity_sha256"
                            ],

                        "membership_status":
                            item[
                                "membership_status"
                            ],
                    },
                    ensure_ascii=False,
                    sort_keys=True,
                )
            )


print(
    "\n=== DETERMINISM AUDIT ==="
)

print(
    "Recomputation failures:",
    len(
        deterministic_recheck_failures
    )
)

print(
    "Selected duplicate identity failures:",
    len(
        selected_duplicate_identity_failures
    )
)


print(
    "\n=== CODE CELL 22 GATES ==="
)

for name, passed in (
    cell_22_gates.items()
):

    print(
        name,
        ":",
        "PASS"
        if passed
        else "FAIL"
    )


print(
    "\nCODE CELL 22 GATE:",
    "PASS"
    if cell_22_pass
    else "FAIL"
)


assert cell_22_pass, (
    "Deterministic Golden candidate selection failed. "
    "Do not proceed to case construction or question design "
    "until quota, diversity, identity, and determinism gates pass."
)
```

<div style="direction: rtl; white-space: normal; line-height: 1.7;">

<h3>Golden Case Record Contract and Stable Case IDs</h3>

<p>
Deterministic Selection اکنون انجام شده است، اما Selected Membership هنوز به‌تنهایی یک Golden Case کامل نیست.
برای ورود به مراحل بعد باید هر Membership به یک Record استاندارد، قابل ردیابی و دارای Stable ID تبدیل شود.
</p>

<p>
در این مرحله برای هر یک از ۷۲۴ Membership انتخاب‌شده یک
<code>golden_case_id</code>
قطعی ساخته می‌شود.
این ID از موارد زیر مشتق می‌شود:
</p>

<ul>
<li>نسخه‌ی Case Contract؛</li>
<li>Source Type؛</li>
<li>Stratum؛</li>
<li>Stable Candidate Identity Hash.</li>
</ul>

<p>
وضعیت Review یا Final Membership عمداً داخل Case ID قرار نمی‌گیرد،
زیرا این وضعیت ممکن است در آینده پس از Human Review تغییر کند و نباید باعث تغییر شناسه‌ی Case شود.
</p>

<p>
همچنین Evidence واقعی هر Case دوباره به Accepted Corpus متصل می‌شود.
برای هر Evidence Document باید Source ID، Locator، Lineage، Attribution و Rights
بدون استنتاج جدید قابل بازیابی باشند.
</p>

<p>
برای Unitهای چندرکوردی مانند Book Section، Protected Region، Transcript Physical File
یا Transition، تمام Document IDهای تشکیل‌دهنده‌ی Evidence ثبت می‌شوند.
</p>

<p>
Review Shortlistها نیز Case ID دریافت می‌کنند، اما همچنان
<code>final_golden_membership = false</code>
دارند و به Golden نهایی تبدیل نمی‌شوند.
</p>

<p>
در این مرحله هنوز هیچ Question، Expected Answer، Citation Target یا Retrieval Judgment تولید نمی‌شود.
</p>

</div>

```python
# Code Cell 23

# ============================================================
# Golden Case Record Contract + Stable Case IDs
#
# Builds traceable case records for all selected memberships.
#
# NO question generation.
# NO answer generation.
# NO artifact writes.
# ============================================================


# ============================================================
# 1) Build full accepted-corpus evidence index
# ============================================================

corpus_evidence_index = {}

transcript_documents_by_physical_file = defaultdict(
    list
)


with corpus_path.open(
    "r",
    encoding="utf-8",
) as handle:

    for line in handle:

        if not line.strip():
            continue

        record = json.loads(
            line
        )

        document_id = (
            record.get(
                "document_id"
            )
        )

        source_type = (
            record.get(
                "source_type"
            )
        )

        locator = (
            record.get(
                "locator"
            )
            or {}
        )

        corpus_evidence_index[
            document_id
        ] = {
            "document_id":
                document_id,

            "source_id":
                record.get(
                    "source_id"
                ),

            "source_type":
                source_type,

            "title":
                record.get(
                    "title"
                ),

            "text_sha256":
                record.get(
                    "text_sha256"
                ),

            "raw_content_hash":
                record.get(
                    "raw_content_hash"
                ),

            "locator":
                locator,

            "lineage":
                record.get(
                    "lineage"
                ),

            "attribution":
                record.get(
                    "attribution"
                ),

            "rights":
                record.get(
                    "rights"
                ),

            "content_origin":
                record.get(
                    "content_origin"
                ),
        }


        if (
            source_type
            == "transcript"
        ):

            physical_file_id = (
                locator.get(
                    "physical_file_id"
                )
            )

            if (
                physical_file_id
                is not None
            ):

                transcript_documents_by_physical_file[
                    physical_file_id
                ].append(
                    document_id
                )


# deterministic order

for physical_file_id in list(
    transcript_documents_by_physical_file
):

    transcript_documents_by_physical_file[
        physical_file_id
    ] = sorted(
        set(
            transcript_documents_by_physical_file[
                physical_file_id
            ]
        )
    )


# ============================================================
# 2) Resolve evidence documents for any selected candidate
# ============================================================

def resolve_case_evidence_document_ids(
    source_type,
    stratum,
    candidate,
):

    evidence_ids = []


    # --------------------------------------------------------
    # Explicit document collections first
    # --------------------------------------------------------

    for field in [
        "document_ids",
        "member_document_ids",
    ]:

        values = (
            candidate.get(
                field
            )
            or []
        )

        if isinstance(
            values,
            list,
        ):

            evidence_ids.extend(
                value
                for value
                in values
                if value is not None
            )


    # --------------------------------------------------------
    # Direct single-document identities
    # --------------------------------------------------------

    for field in [
        "document_id",
        "left_document_id",
        "right_document_id",
        "parent_paragraph_id",
    ]:

        value = (
            candidate.get(
                field
            )
        )

        if (
            value is not None
            and
            value
            in corpus_evidence_index
        ):

            evidence_ids.append(
                value
            )


    # --------------------------------------------------------
    # Physical-file cases
    # --------------------------------------------------------

    physical_file_id = (
        candidate.get(
            "physical_file_id"
        )
    )

    if (
        physical_file_id
        is not None
        and
        stratum
        in {
            "single_speaker_primary",
            "single_speaker_unresolved",
            "multi_speaker_uni_tehran",
            "multi_speaker_chehelston",
            "multi_speaker_asre_shirin",
            "multi_speaker_khane_to",
            "multi_speaker_mizan",
            "mixed_primary_signal_case",
        }
    ):

        evidence_ids.extend(
            transcript_documents_by_physical_file.get(
                physical_file_id,
                [],
            )
        )


    # --------------------------------------------------------
    # Keep only accepted-corpus identities.
    # No fabricated evidence identifiers.
    # --------------------------------------------------------

    evidence_ids = sorted(
        {
            document_id
            for document_id
            in evidence_ids
            if (
                document_id
                in corpus_evidence_index
            )
        }
    )

    return evidence_ids


# ============================================================
# 3) Stable Golden Case ID
# ============================================================

GOLDEN_CASE_SCHEMA_VERSION = (
    "golden_case_record_v1_draft"
)


def build_golden_case_id(
    source_type,
    stratum,
    candidate_identity_sha256,
):

    case_identity_payload = {
        "case_schema_version":
            GOLDEN_CASE_SCHEMA_VERSION,

        "source_type":
            source_type,

        "stratum":
            stratum,

        "candidate_identity_sha256":
            candidate_identity_sha256,
    }

    canonical = stable_json_key(
        case_identity_payload
    )

    digest = hashlib.sha256(
        canonical.encode(
            "utf-8"
        )
    ).hexdigest()

    return (
        "golden-case:"
        + digest[
            :32
        ]
    )


# ============================================================
# 4) Case Record Contract
# ============================================================

GOLDEN_CASE_RECORD_CONTRACT_V1_DRAFT = {
    "version":
        GOLDEN_CASE_SCHEMA_VERSION,

    "status":
        "DRAFT_VALIDATED_NOT_FROZEN",

    "case_identity_policy": {
        "algorithm":
            "SHA256",

        "inputs": [
            "case_schema_version",
            "source_type",
            "stratum",
            "candidate_identity_sha256",
        ],

        "mutable_membership_status_excluded":
            True,

        "randomness":
            False,
    },

    "required_fields": [
        "golden_case_id",
        "case_schema_version",
        "source_type",
        "stratum",
        "selection_rank",
        "candidate_identity",
        "candidate_identity_sha256",
        "membership_status",
        "final_golden_membership",
        "evidence_unit",
        "evidence_document_ids",
        "evidence_references",
        "question_generation_status",
    ],

    "evidence_reference_required_fields": [
        "document_id",
        "source_id",
        "source_type",
        "locator",
        "lineage",
        "attribution",
        "rights",
    ],

    "attribution_policy":
        "PROPAGATE_ONLY_FROM_ACCEPTED_CORPUS",

    "rights_policy":
        "PROPAGATE_ONLY_NO_PERMISSION_INFERENCE",

    "unknown_speaker_inference":
        "FORBIDDEN",

    "review_membership_policy":
        (
            "REVIEW_SHORTLIST_CASES_KEEP_STABLE_CASE_ID_"
            "BUT_ARE_NOT_FINAL_GOLDEN_BEFORE_HUMAN_CONFIRMATION"
        ),

    "question_generation_status":
        "NOT_STARTED",

    "expected_answer_generation_status":
        "NOT_STARTED",

    "artifact_write_status":
        "NONE",
}


# ============================================================
# 5) Materialize case records
# ============================================================

golden_case_records = []

golden_case_records_by_id = {}

case_resolution_failures = []

case_source_type_mismatches = []


for source_type in [
    "website",
    "book",
    "transcript",
]:

    for stratum in (
        expected_strata[
            source_type
        ]
    ):

        selected = (
            selected_candidates[
                source_type
            ][
                stratum
            ]
        )

        evidence_unit = (
            STRATUM_RULES_V1_DRAFT[
                source_type
            ][
                stratum
            ].get(
                "unit"
            )
        )


        for selected_item in selected:

            candidate = (
                selected_item[
                    "candidate"
                ]
            )

            evidence_document_ids = (
                resolve_case_evidence_document_ids(
                    source_type,
                    stratum,
                    candidate,
                )
            )


            if not evidence_document_ids:

                case_resolution_failures.append(
                    {
                        "source_type":
                            source_type,

                        "stratum":
                            stratum,

                        "candidate_identity":
                            selected_item[
                                "candidate_identity"
                            ],
                    }
                )


            evidence_references = []


            for document_id in (
                evidence_document_ids
            ):

                evidence_record = (
                    corpus_evidence_index[
                        document_id
                    ]
                )

                if (
                    evidence_record[
                        "source_type"
                    ]
                    != source_type
                ):

                    case_source_type_mismatches.append(
                        {
                            "source_type":
                                source_type,

                            "stratum":
                                stratum,

                            "document_id":
                                document_id,

                            "document_source_type":
                                evidence_record[
                                    "source_type"
                                ],
                        }
                    )


                evidence_references.append(
                    {
                        "document_id":
                            evidence_record[
                                "document_id"
                            ],

                        "source_id":
                            evidence_record[
                                "source_id"
                            ],

                        "source_type":
                            evidence_record[
                                "source_type"
                            ],

                        "title":
                            evidence_record[
                                "title"
                            ],

                        "text_sha256":
                            evidence_record[
                                "text_sha256"
                            ],

                        "raw_content_hash":
                            evidence_record[
                                "raw_content_hash"
                            ],

                        "locator":
                            evidence_record[
                                "locator"
                            ],

                        "lineage":
                            evidence_record[
                                "lineage"
                            ],

                        "attribution":
                            evidence_record[
                                "attribution"
                            ],

                        "rights":
                            evidence_record[
                                "rights"
                            ],

                        "content_origin":
                            evidence_record[
                                "content_origin"
                            ],
                    }
                )


            golden_case_id = (
                build_golden_case_id(
                    source_type=(
                        source_type
                    ),
                    stratum=(
                        stratum
                    ),
                    candidate_identity_sha256=(
                        selected_item[
                            "candidate_identity_sha256"
                        ]
                    ),
                )
            )


            case_record = {
                "golden_case_id":
                    golden_case_id,

                "case_schema_version":
                    GOLDEN_CASE_SCHEMA_VERSION,

                "source_type":
                    source_type,

                "stratum":
                    stratum,

                "evidence_unit":
                    evidence_unit,

                "selection_rank":
                    selected_item[
                        "selection_rank"
                    ],

                "candidate_identity":
                    selected_item[
                        "candidate_identity"
                    ],

                "candidate_identity_sha256":
                    selected_item[
                        "candidate_identity_sha256"
                    ],

                "membership_status":
                    selected_item[
                        "membership_status"
                    ],

                "final_golden_membership":
                    selected_item[
                        "final_golden_membership"
                    ],

                "evidence_document_ids":
                    evidence_document_ids,

                "evidence_document_count":
                    len(
                        evidence_document_ids
                    ),

                "evidence_references":
                    evidence_references,

                "candidate_snapshot":
                    candidate,

                "selection_provenance": {
                    "sampling_policy_version":
                        SAMPLING_POLICY_V1_DRAFT[
                            "version"
                        ],

                    "selection_method":
                        GOLDEN_SELECTION_V1_DRAFT[
                            "selection_method"
                        ],

                    "sampling_target":
                        SAMPLING_TARGET,

                    "random_sampling":
                        False,

                    "random_seed":
                        None,
                },

                "question_generation_status":
                    "NOT_STARTED",

                "expected_answer_generation_status":
                    "NOT_STARTED",

                "review_status":
                    (
                        "NOT_REQUIRED"
                        if (
                            selected_item[
                                "final_golden_membership"
                            ]
                            is True
                        )
                        else
                        "PENDING_HUMAN_CONFIRMATION"
                    ),
            }


            golden_case_records.append(
                case_record
            )


            if (
                golden_case_id
                in golden_case_records_by_id
            ):

                raise AssertionError(
                    "Duplicate golden_case_id generated: "
                    f"{golden_case_id}"
                )


            golden_case_records_by_id[
                golden_case_id
            ] = (
                case_record
            )


# ============================================================
# 6) Independent stable-ID recomputation
# ============================================================

case_id_recomputation_failures = []


for case_record in (
    golden_case_records
):

    recomputed_id = (
        build_golden_case_id(
            source_type=(
                case_record[
                    "source_type"
                ]
            ),
            stratum=(
                case_record[
                    "stratum"
                ]
            ),
            candidate_identity_sha256=(
                case_record[
                    "candidate_identity_sha256"
                ]
            ),
        )
    )


    if (
        recomputed_id
        !=
        case_record[
            "golden_case_id"
        ]
    ):

        case_id_recomputation_failures.append(
            case_record[
                "golden_case_id"
            ]
        )


# ============================================================
# 7) Evidence integrity audit
# ============================================================

missing_required_case_fields = []

missing_evidence_reference_fields = []

locator_missing_cases = []

lineage_missing_cases = []

rights_missing_cases = []

attribution_missing_cases = []


required_case_fields = (
    GOLDEN_CASE_RECORD_CONTRACT_V1_DRAFT[
        "required_fields"
    ]
)


required_evidence_fields = (
    GOLDEN_CASE_RECORD_CONTRACT_V1_DRAFT[
        "evidence_reference_required_fields"
    ]
)


for case_record in (
    golden_case_records
):

    missing_case_fields = [
        field
        for field
        in required_case_fields
        if (
            field
            not in case_record
        )
    ]

    if missing_case_fields:

        missing_required_case_fields.append(
            {
                "golden_case_id":
                    case_record[
                        "golden_case_id"
                    ],

                "missing_fields":
                    missing_case_fields,
            }
        )


    for evidence_reference in (
        case_record[
            "evidence_references"
        ]
    ):

        missing_fields = [
            field
            for field
            in required_evidence_fields
            if (
                field
                not in evidence_reference
            )
        ]

        if missing_fields:

            missing_evidence_reference_fields.append(
                {
                    "golden_case_id":
                        case_record[
                            "golden_case_id"
                        ],

                    "document_id":
                        evidence_reference[
                            "document_id"
                        ],

                    "missing_fields":
                        missing_fields,
                }
            )


        if not evidence_reference.get(
            "locator"
        ):

            locator_missing_cases.append(
                (
                    case_record[
                        "golden_case_id"
                    ],
                    evidence_reference[
                        "document_id"
                    ],
                )
            )


        if not evidence_reference.get(
            "lineage"
        ):

            lineage_missing_cases.append(
                (
                    case_record[
                        "golden_case_id"
                    ],
                    evidence_reference[
                        "document_id"
                    ],
                )
            )


        if (
            evidence_reference.get(
                "rights"
            )
            is None
        ):

            rights_missing_cases.append(
                (
                    case_record[
                        "golden_case_id"
                    ],
                    evidence_reference[
                        "document_id"
                    ],
                )
            )


        if (
            evidence_reference.get(
                "attribution"
            )
            is None
        ):

            attribution_missing_cases.append(
                (
                    case_record[
                        "golden_case_id"
                    ],
                    evidence_reference[
                        "document_id"
                    ],
                )
            )


# ============================================================
# 8) Membership accounting
# ============================================================

auto_case_records = [
    record
    for record
    in golden_case_records
    if (
        record[
            "final_golden_membership"
        ]
        is True
    )
]


review_case_records = [
    record
    for record
    in golden_case_records
    if (
        record[
            "final_golden_membership"
        ]
        is False
    )
]


case_counts_by_source = Counter(
    record[
        "source_type"
    ]
    for record
    in golden_case_records
)


case_counts_by_stratum = Counter(
    (
        record[
            "source_type"
        ],
        record[
            "stratum"
        ],
    )
    for record
    in golden_case_records
)


# ============================================================
# 9) Cross-stratum repeated candidate identity audit
#
# Repetition across strata is allowed and intentionally retained.
# ============================================================

candidate_hash_to_case_ids = defaultdict(
    list
)


for case_record in (
    golden_case_records
):

    candidate_hash_to_case_ids[
        case_record[
            "candidate_identity_sha256"
        ]
    ].append(
        case_record[
            "golden_case_id"
        ]
    )


cross_stratum_repeated_candidate_hashes = {
    candidate_hash:
        case_ids

    for candidate_hash, case_ids
    in candidate_hash_to_case_ids.items()

    if (
        len(
            case_ids
        )
        > 1
    )
}


# ============================================================
# 10) Golden Case Registry draft
# ============================================================

GOLDEN_CASE_REGISTRY_V1_DRAFT = {
    "version":
        "golden_case_registry_v1_draft",

    "status":
        "MATERIALIZED_IN_MEMORY_NOT_FROZEN",

    "case_contract_version":
        GOLDEN_CASE_SCHEMA_VERSION,

    "selected_membership_count":
        len(
            golden_case_records
        ),

    "auto_golden_case_count":
        len(
            auto_case_records
        ),

    "review_pending_case_count":
        len(
            review_case_records
        ),

    "source_counts":
        dict(
            case_counts_by_source
        ),

    "cross_stratum_repeated_candidate_hash_count":
        len(
            cross_stratum_repeated_candidate_hashes
        ),

    "cross_stratum_policy":
        "PRESERVED_NOT_DEDUPLICATED",

    "question_generation_status":
        "NOT_STARTED",

    "expected_answer_generation_status":
        "NOT_STARTED",

    "artifact_write_status":
        "NONE",
}


# ============================================================
# 11) Validation gates
# ============================================================

selected_membership_count_gate = (
    len(
        golden_case_records
    )
    ==
    selected_total_memberships
    ==
    724
)


auto_case_count_gate = (
    len(
        auto_case_records
    )
    ==
    selected_auto_memberships
    ==
    572
)


review_case_count_gate = (
    len(
        review_case_records
    )
    ==
    selected_review_shortlist_memberships
    ==
    152
)


case_ids_unique_gate = (
    len(
        golden_case_records_by_id
    )
    ==
    len(
        golden_case_records
    )
)


case_ids_recomputable_gate = (
    len(
        case_id_recomputation_failures
    )
    == 0
)


all_cases_resolve_evidence_gate = (
    len(
        case_resolution_failures
    )
    == 0
)


source_type_integrity_gate = (
    len(
        case_source_type_mismatches
    )
    == 0
)


required_case_fields_gate = (
    len(
        missing_required_case_fields
    )
    == 0
)


required_evidence_fields_gate = (
    len(
        missing_evidence_reference_fields
    )
    == 0
)


locator_recoverable_gate = (
    len(
        locator_missing_cases
    )
    == 0
)


lineage_recoverable_gate = (
    len(
        lineage_missing_cases
    )
    == 0
)


rights_propagated_gate = (
    len(
        rights_missing_cases
    )
    == 0
)


attribution_propagated_gate = (
    len(
        attribution_missing_cases
    )
    == 0
)


review_cases_not_final_gate = all(
    record[
        "review_status"
    ]
    ==
    "PENDING_HUMAN_CONFIRMATION"

    and

    record[
        "final_golden_membership"
    ]
    is False

    for record
    in review_case_records
)


auto_cases_final_gate = all(
    record[
        "review_status"
    ]
    ==
    "NOT_REQUIRED"

    and

    record[
        "final_golden_membership"
    ]
    is True

    for record
    in auto_case_records
)


question_generation_not_started_gate = all(
    record[
        "question_generation_status"
    ]
    == "NOT_STARTED"

    and

    record[
        "expected_answer_generation_status"
    ]
    == "NOT_STARTED"

    for record
    in golden_case_records
)


no_artifact_write_gate = (
    GOLDEN_CASE_REGISTRY_V1_DRAFT[
        "artifact_write_status"
    ]
    == "NONE"

    and

    GOLDEN_CASE_RECORD_CONTRACT_V1_DRAFT[
        "artifact_write_status"
    ]
    == "NONE"
)


cell_23_gates = {
    "exact_724_selected_memberships_materialized":
        selected_membership_count_gate,

    "exact_572_auto_golden_cases":
        auto_case_count_gate,

    "exact_152_review_pending_cases":
        review_case_count_gate,

    "golden_case_ids_unique":
        case_ids_unique_gate,

    "golden_case_ids_independently_recomputable":
        case_ids_recomputable_gate,

    "all_cases_resolve_to_accepted_corpus_evidence":
        all_cases_resolve_evidence_gate,

    "evidence_source_type_integrity":
        source_type_integrity_gate,

    "all_required_case_fields_present":
        required_case_fields_gate,

    "all_required_evidence_reference_fields_present":
        required_evidence_fields_gate,

    "locator_recoverability":
        locator_recoverable_gate,

    "lineage_recoverability":
        lineage_recoverable_gate,

    "rights_propagated_without_inference":
        rights_propagated_gate,

    "attribution_propagated_without_inference":
        attribution_propagated_gate,

    "review_cases_remain_nonfinal":
        review_cases_not_final_gate,

    "auto_cases_are_final_memberships":
        auto_cases_final_gate,

    "question_and_answer_generation_not_started":
        question_generation_not_started_gate,

    "no_artifact_write":
        no_artifact_write_gate,
}


cell_23_pass = all(
    cell_23_gates.values()
)


# ============================================================
# 12) Output
# ============================================================

print(
    "=== GOLDEN CASE REGISTRY ==="
)

print(
    "Total selected case memberships:",
    len(
        golden_case_records
    )
)

print(
    "Auto Golden cases:",
    len(
        auto_case_records
    )
)

print(
    "Review-pending cases:",
    len(
        review_case_records
    )
)

print(
    "Unique Golden Case IDs:",
    len(
        golden_case_records_by_id
    )
)


print(
    "\n=== CASE COUNTS BY SOURCE ==="
)

for source_type in [
    "website",
    "book",
    "transcript",
]:

    print(
        "-",
        source_type,
        ":",
        case_counts_by_source[
            source_type
        ],
    )


print(
    "\n=== CROSS-STRATUM REPEATED CANDIDATE IDENTITIES ==="
)

print(
    "Repeated candidate hashes:",
    len(
        cross_stratum_repeated_candidate_hashes
    )
)

print(
    "Policy:",
    "PRESERVED_NOT_DEDUPLICATED"
)


print(
    "\n=== REPRESENTATIVE GOLDEN CASE RECORDS ==="
)

for record in (
    golden_case_records[
        :5
    ]
):

    print(
        json.dumps(
            {
                "golden_case_id":
                    record[
                        "golden_case_id"
                    ],

                "source_type":
                    record[
                        "source_type"
                    ],

                "stratum":
                    record[
                        "stratum"
                    ],

                "selection_rank":
                    record[
                        "selection_rank"
                    ],

                "membership_status":
                    record[
                        "membership_status"
                    ],

                "final_golden_membership":
                    record[
                        "final_golden_membership"
                    ],

                "evidence_unit":
                    record[
                        "evidence_unit"
                    ],

                "evidence_document_count":
                    record[
                        "evidence_document_count"
                    ],

                "evidence_document_ids_preview":
                    record[
                        "evidence_document_ids"
                    ][:5],

                "question_generation_status":
                    record[
                        "question_generation_status"
                    ],
            },
            ensure_ascii=False,
            sort_keys=True,
        )
    )


print(
    "\n=== EVIDENCE INTEGRITY FAILURES ==="
)

print(
    "Case resolution failures:",
    len(
        case_resolution_failures
    )
)

print(
    "Source-type mismatches:",
    len(
        case_source_type_mismatches
    )
)

print(
    "Missing required case fields:",
    len(
        missing_required_case_fields
    )
)

print(
    "Missing evidence fields:",
    len(
        missing_evidence_reference_fields
    )
)

print(
    "Missing locator:",
    len(
        locator_missing_cases
    )
)

print(
    "Missing lineage:",
    len(
        lineage_missing_cases
    )
)

print(
    "Missing rights:",
    len(
        rights_missing_cases
    )
)

print(
    "Missing attribution:",
    len(
        attribution_missing_cases
    )
)


print(
    "\n=== CODE CELL 23 GATES ==="
)

for name, passed in (
    cell_23_gates.items()
):

    print(
        name,
        ":",
        "PASS"
        if passed
        else "FAIL"
    )


print(
    "\nCODE CELL 23 GATE:",
    "PASS"
    if cell_23_pass
    else "FAIL"
)


assert cell_23_pass, (
    "Golden Case Record construction failed. "
    "Do not proceed to question/evidence-task design until "
    "case IDs, corpus linkage, locator, lineage, attribution, "
    "and rights integrity all pass."
)
```

<h3>Evaluation Task Design Contract</h3>

<p>
Golden Case Registry اکنون دارای Case ID، Evidence، Locator، Lineage، Attribution و Rights معتبر است.
اما هنوز مشخص نشده هر Stratum دقیقاً چه نوع رفتار RAG را باید ارزیابی کند.
</p>

<p>
در این مرحله برای تمام ۲۷ Stratum یک Task Design Contract تعریف می‌شود.
این Contract سؤال واقعی تولید نمی‌کند؛ فقط تعیین می‌کند هر Case در آینده چه قابلیت‌هایی را باید آزمایش کند.
</p>

<p>
Evaluation Dimensions فریز‌شده‌ی NB11/NB12 عبارت‌اند از:
</p>

<ul>
<li><code>retrieval_hit</code></li>
<li><code>rank</code></li>
<li><code>evidence_completeness</code></li>
<li><code>attribution_correctness</code></li>
<li><code>citation_locator_correctness</code></li>
<li><code>unnecessary_context</code></li>
<li><code>source_authority_correctness</code></li>
</ul>

<p>
هر Stratum یک یا چند Dimension اصلی دارد، اما هیچ Dimension جدیدی خارج از Contract ایجاد نمی‌شود.
</p>

<p>
چهار Stratum دارای Human Review هنوز Task قابل اجرا دریافت نمی‌کنند.
برای آن‌ها فقط Design ثبت می‌شود و وضعیت
<code>BLOCKED_PENDING_HUMAN_CONFIRMATION</code>
حفظ می‌شود.
</p>

<p>
Unknown Speaker Inference همچنان ممنوع است.
همچنین Secondary یا Contextual Source نباید به‌صورت خودکار به Primary Authority ارتقا داده شود.
</p>

<p>
در این مرحله هیچ Question، Expected Answer، Retrieval Judgment، Citation Target یا Artifact نوشته نمی‌شود.
</p>

</div>

```python
# Code Cell 24

# ============================================================
# Evaluation Task Design Contract
#
# Defines task archetypes only.
#
# NO question generation.
# NO answer generation.
# NO retrieval execution.
# NO artifact writes.
# ============================================================


FROZEN_EVALUATION_DIMENSIONS = [
    "retrieval_hit",
    "rank",
    "evidence_completeness",
    "attribution_correctness",
    "citation_locator_correctness",
    "unnecessary_context",
    "source_authority_correctness",
]


# ============================================================
# 1) Task archetype registry
# ============================================================

TASK_DESIGN_BY_STRATUM = {

    # ========================================================
    # WEBSITE
    # ========================================================

    "website": {

        "direct_author_short": {
            "task_family":
                "DIRECT_PRIMARY_SOURCE_RETRIEVAL",

            "evaluation_goal":
                (
                    "Retrieve a short Website document directly "
                    "attributed to Dr. Komeil Roudi without replacing "
                    "it with supervised or contextual evidence."
                ),

            "evidence_scope":
                "SINGLE_SOURCE_DOCUMENT",

            "required_dimensions": [
                "retrieval_hit",
                "rank",
                "attribution_correctness",
                "citation_locator_correctness",
                "source_authority_correctness",
            ],

            "context_expectation":
                "MINIMAL_DIRECT_EVIDENCE",

            "authority_expectation":
                "PRIMARY_DIRECT_AUTHOR",
        },


        "direct_author_long": {
            "task_family":
                "LONG_DIRECT_PRIMARY_SOURCE_RETRIEVAL",

            "evaluation_goal":
                (
                    "Retrieve sufficient evidence from a long Website "
                    "document directly attributed to Dr. Komeil Roudi."
                ),

            "evidence_scope":
                "SINGLE_LONG_SOURCE_DOCUMENT",

            "required_dimensions": [
                "retrieval_hit",
                "rank",
                "evidence_completeness",
                "attribution_correctness",
                "citation_locator_correctness",
                "unnecessary_context",
                "source_authority_correctness",
            ],

            "context_expectation":
                "SUFFICIENT_WITHOUT_EXCESS",

            "authority_expectation":
                "PRIMARY_DIRECT_AUTHOR",
        },


        "supervised_short": {
            "task_family":
                "SUPERVISED_SECONDARY_SOURCE_RETRIEVAL",

            "evaluation_goal":
                (
                    "Retrieve supervised Website evidence while keeping "
                    "its authority distinct from direct Dr. Roudi authorship."
                ),

            "evidence_scope":
                "SINGLE_SOURCE_DOCUMENT",

            "required_dimensions": [
                "retrieval_hit",
                "rank",
                "attribution_correctness",
                "citation_locator_correctness",
                "source_authority_correctness",
            ],

            "context_expectation":
                "MINIMAL_SECONDARY_EVIDENCE",

            "authority_expectation":
                "SUPERVISED_SECONDARY",
        },


        "supervised_long": {
            "task_family":
                "LONG_SUPERVISED_SECONDARY_SOURCE_RETRIEVAL",

            "evaluation_goal":
                (
                    "Retrieve sufficient evidence from a long supervised "
                    "Website article without promoting it to direct authorship."
                ),

            "evidence_scope":
                "SINGLE_LONG_SOURCE_DOCUMENT",

            "required_dimensions": [
                "retrieval_hit",
                "rank",
                "evidence_completeness",
                "attribution_correctness",
                "citation_locator_correctness",
                "unnecessary_context",
                "source_authority_correctness",
            ],

            "context_expectation":
                "SUFFICIENT_WITHOUT_EXCESS",

            "authority_expectation":
                "SUPERVISED_SECONDARY",
        },


        "multi_topic_article": {
            "task_family":
                "MULTI_TOPIC_EVIDENCE_DISAMBIGUATION",

            "evaluation_goal":
                (
                    "Identify the correct local evidence inside a potentially "
                    "multi-topic Website article without retrieving unrelated "
                    "sections."
                ),

            "evidence_scope":
                "LOCAL_EVIDENCE_WITHIN_SOURCE_DOCUMENT",

            "required_dimensions": [
                "retrieval_hit",
                "rank",
                "evidence_completeness",
                "citation_locator_correctness",
                "unnecessary_context",
            ],

            "context_expectation":
                "TOPIC_LOCALIZED_CONTEXT",

            "authority_expectation":
                "PRESERVE_SOURCE_AUTHORITY",
        },


        "citation_locator_sensitive_case": {
            "task_family":
                "CITATION_LOCATOR_DISAMBIGUATION",

            "evaluation_goal":
                (
                    "Retrieve the correct Website evidence when semantically "
                    "similar material exists elsewhere and preserve the exact "
                    "source locator."
                ),

            "evidence_scope":
                "SOURCE_DOCUMENT_PLUS_SIMILARITY_CONTRAST",

            "required_dimensions": [
                "retrieval_hit",
                "rank",
                "citation_locator_correctness",
                "attribution_correctness",
                "source_authority_correctness",
            ],

            "context_expectation":
                "CORRECT_SOURCE_NOT_SEMANTIC_LOOKALIKE",

            "authority_expectation":
                "PRESERVE_EXACT_SOURCE_IDENTITY",
        },
    },


    # ========================================================
    # BOOK
    # ========================================================

    "book": {

        "intrinsically_tiny_narrative_region": {
            "task_family":
                "TINY_NARRATIVE_EVIDENCE_RETRIEVAL",

            "evaluation_goal":
                (
                    "Retrieve intrinsically small narrative evidence without "
                    "losing it or incorrectly merging it with unrelated regions."
                ),

            "evidence_scope":
                "PROTECTED_STRUCTURAL_REGION",

            "required_dimensions": [
                "retrieval_hit",
                "rank",
                "evidence_completeness",
                "citation_locator_correctness",
                "unnecessary_context",
            ],

            "context_expectation":
                "PRESERVE_SMALL_REGION",

            "authority_expectation":
                "BOOK_PRIMARY_EVIDENCE",
        },


        "intrinsically_tiny_table_region": {
            "task_family":
                "TINY_TABLE_EVIDENCE_RETRIEVAL",

            "evaluation_goal":
                (
                    "Retrieve a small protected table region while preserving "
                    "its structural and locator boundaries."
                ),

            "evidence_scope":
                "PROTECTED_TABLE_REGION",

            "required_dimensions": [
                "retrieval_hit",
                "rank",
                "evidence_completeness",
                "citation_locator_correctness",
                "unnecessary_context",
            ],

            "context_expectation":
                "PRESERVE_TABLE_REGION",

            "authority_expectation":
                "BOOK_PRIMARY_EVIDENCE",
        },


        "long_section": {
            "task_family":
                "LONG_SECTION_EVIDENCE_COMPLETENESS",

            "evaluation_goal":
                (
                    "Retrieve enough evidence from a long Book section when "
                    "the answer may require more than one retrieval chunk."
                ),

            "evidence_scope":
                "BOOK_SECTION",

            "required_dimensions": [
                "retrieval_hit",
                "rank",
                "evidence_completeness",
                "citation_locator_correctness",
                "unnecessary_context",
            ],

            "context_expectation":
                "MULTI_CHUNK_IF_REQUIRED",

            "authority_expectation":
                "BOOK_PRIMARY_EVIDENCE",
        },


        "large_table_run": {
            "task_family":
                "LARGE_TABLE_RUN_RETRIEVAL",

            "evaluation_goal":
                (
                    "Retrieve evidence spanning a large contiguous table run "
                    "without breaking the structural meaning of the table."
                ),

            "evidence_scope":
                "CONTIGUOUS_TABLE_RUN",

            "required_dimensions": [
                "retrieval_hit",
                "rank",
                "evidence_completeness",
                "citation_locator_correctness",
                "unnecessary_context",
            ],

            "context_expectation":
                "TABLE_STRUCTURE_PRESERVED",

            "authority_expectation":
                "BOOK_PRIMARY_EVIDENCE",
        },


        "exercise_or_self_assessment": {
            "task_family":
                "EXERCISE_SELF_ASSESSMENT_RETRIEVAL",

            "evaluation_goal":
                (
                    "Retrieve exercise or self-assessment material as such, "
                    "without confusing instructional prompts with explanatory prose."
                ),

            "evidence_scope":
                "BOOK_REGION",

            "required_dimensions": [
                "retrieval_hit",
                "rank",
                "evidence_completeness",
                "citation_locator_correctness",
                "unnecessary_context",
            ],

            "context_expectation":
                "EXERCISE_BOUNDARY_PRESERVED",

            "authority_expectation":
                "BOOK_PRIMARY_EVIDENCE",
        },


        "supplemental_footnote_case": {
            "task_family":
                "PARENT_SUPPLEMENTAL_EVIDENCE_RETRIEVAL",

            "evaluation_goal":
                (
                    "Retrieve a supplemental footnote together with sufficient "
                    "parent context when the supplemental text alone is incomplete."
                ),

            "evidence_scope":
                "PARENT_PLUS_SUPPLEMENTAL_RECORD",

            "required_dimensions": [
                "retrieval_hit",
                "rank",
                "evidence_completeness",
                "citation_locator_correctness",
                "unnecessary_context",
            ],

            "context_expectation":
                "PARENT_CONTEXT_WHEN_NEEDED",

            "authority_expectation":
                "BOOK_PRIMARY_EVIDENCE",
        },


        "deep_heading_path": {
            "task_family":
                "DEEP_HIERARCHICAL_LOCATOR_RETRIEVAL",

            "evaluation_goal":
                (
                    "Retrieve the correct Book evidence while preserving the "
                    "deep heading hierarchy required for precise citation."
                ),

            "evidence_scope":
                "BOOK_RECORD_WITH_HIERARCHICAL_CONTEXT",

            "required_dimensions": [
                "retrieval_hit",
                "rank",
                "citation_locator_correctness",
                "evidence_completeness",
            ],

            "context_expectation":
                "HEADING_PATH_PRESERVED",

            "authority_expectation":
                "BOOK_PRIMARY_EVIDENCE",
        },


        "multi_paragraph_evidence_case": {
            "task_family":
                "MULTI_PARAGRAPH_EVIDENCE_ASSEMBLY",

            "evaluation_goal":
                (
                    "Determine whether a question genuinely requires evidence "
                    "from multiple paragraphs of the same Book section."
                ),

            "evidence_scope":
                "BOOK_SECTION_MULTI_PARAGRAPH",

            "required_dimensions": [
                "retrieval_hit",
                "rank",
                "evidence_completeness",
                "citation_locator_correctness",
                "unnecessary_context",
            ],

            "context_expectation":
                "MULTI_PARAGRAPH_ONLY_IF_REQUIRED",

            "authority_expectation":
                "BOOK_PRIMARY_EVIDENCE",
        },
    },


    # ========================================================
    # TRANSCRIPT
    # ========================================================

    "transcript": {

        "single_speaker_primary": {
            "task_family":
                "PRIMARY_TRANSCRIPT_RETRIEVAL",

            "evaluation_goal":
                (
                    "Retrieve transcript evidence explicitly attributable "
                    "to Dr. Komeil Roudi."
                ),

            "evidence_scope":
                "PHYSICAL_FILE_LOCAL_EVIDENCE",

            "required_dimensions": [
                "retrieval_hit",
                "rank",
                "attribution_correctness",
                "citation_locator_correctness",
                "source_authority_correctness",
            ],

            "context_expectation":
                "PRIMARY_SPEAKER_CONTEXT",

            "authority_expectation":
                "PRIMARY_DR_ROUDI",
        },


        "single_speaker_unresolved": {
            "task_family":
                "UNRESOLVED_SPEAKER_ABSTENTION",

            "evaluation_goal":
                (
                    "Retrieve unresolved transcript evidence without inferring "
                    "a speaker identity or attributing it to Dr. Roudi."
                ),

            "evidence_scope":
                "PHYSICAL_FILE_LOCAL_EVIDENCE",

            "required_dimensions": [
                "retrieval_hit",
                "attribution_correctness",
                "source_authority_correctness",
                "unnecessary_context",
            ],

            "context_expectation":
                "UNRESOLVED_IDENTITY_PRESERVED",

            "authority_expectation":
                "UNRESOLVED_SPEAKER_CONTEXT",
        },


        "multi_speaker_uni_tehran": {
            "task_family":
                "MULTI_SPEAKER_ATTRIBUTION_UNI_TEHRAN",

            "evaluation_goal":
                (
                    "Retrieve evidence from a multi-speaker Uni Tehran file "
                    "while preserving utterance-level attribution."
                ),

            "evidence_scope":
                "PHYSICAL_FILE_MULTI_SPEAKER",

            "required_dimensions": [
                "retrieval_hit",
                "rank",
                "attribution_correctness",
                "citation_locator_correctness",
                "unnecessary_context",
                "source_authority_correctness",
            ],

            "context_expectation":
                "NO_MULTI_SPEAKER_COLLAPSE",

            "authority_expectation":
                "UTTERANCE_LEVEL_AUTHORITY",
        },


        "multi_speaker_chehelston": {
            "task_family":
                "MULTI_SPEAKER_ATTRIBUTION_CHEHELSTON",

            "evaluation_goal":
                "Preserve speaker identity in Chehelston multi-party context.",

            "evidence_scope":
                "PHYSICAL_FILE_MULTI_SPEAKER",

            "required_dimensions": [
                "retrieval_hit",
                "rank",
                "attribution_correctness",
                "citation_locator_correctness",
                "source_authority_correctness",
            ],

            "context_expectation":
                "NO_MULTI_SPEAKER_COLLAPSE",

            "authority_expectation":
                "UTTERANCE_LEVEL_AUTHORITY",
        },


        "multi_speaker_asre_shirin": {
            "task_family":
                "MULTI_SPEAKER_ATTRIBUTION_ASRE_SHIRIN",

            "evaluation_goal":
                (
                    "Preserve explicit speaker labels without treating "
                    "speaker labels as proof of primary Dr. Roudi authority."
                ),

            "evidence_scope":
                "PHYSICAL_FILE_MULTI_SPEAKER",

            "required_dimensions": [
                "retrieval_hit",
                "attribution_correctness",
                "source_authority_correctness",
                "unnecessary_context",
            ],

            "context_expectation":
                "EXPLICIT_LABELS_PRESERVED",

            "authority_expectation":
                "DO_NOT_INFER_PRIMARY_FROM_LABEL",
        },


        "multi_speaker_khane_to": {
            "task_family":
                "MULTI_SPEAKER_ATTRIBUTION_KHANE_TO",

            "evaluation_goal":
                (
                    "Separate Dr. Roudi and host/other speaker evidence "
                    "inside Khane-to transcript context."
                ),

            "evidence_scope":
                "PHYSICAL_FILE_MULTI_SPEAKER",

            "required_dimensions": [
                "retrieval_hit",
                "rank",
                "attribution_correctness",
                "source_authority_correctness",
            ],

            "context_expectation":
                "SPEAKER_BOUNDARIES_PRESERVED",

            "authority_expectation":
                "UTTERANCE_LEVEL_AUTHORITY",
        },


        "multi_speaker_mizan": {
            "task_family":
                "MULTI_SPEAKER_ATTRIBUTION_MIZAN",

            "evaluation_goal":
                (
                    "Evaluate the single available Mizan multi-speaker file "
                    "without inferring unresolved speaker identity."
                ),

            "evidence_scope":
                "PHYSICAL_FILE_MULTI_SPEAKER",

            "required_dimensions": [
                "retrieval_hit",
                "attribution_correctness",
                "source_authority_correctness",
            ],

            "context_expectation":
                "UNRESOLVED_IDENTITY_PRESERVED",

            "authority_expectation":
                "UTTERANCE_LEVEL_AUTHORITY",
        },


        "primary_to_nonprimary_transition": {
            "task_family":
                "PRIMARY_NONPRIMARY_BOUNDARY_RETRIEVAL",

            "evaluation_goal":
                (
                    "Retrieve evidence around a verified Dr. Roudi to explicit "
                    "non-primary speaker boundary without blending attribution."
                ),

            "evidence_scope":
                "ADJACENT_UTTERANCE_PAIR",

            "required_dimensions": [
                "retrieval_hit",
                "rank",
                "attribution_correctness",
                "citation_locator_correctness",
                "unnecessary_context",
                "source_authority_correctness",
            ],

            "context_expectation":
                "SPEAKER_TRANSITION_PRESERVED",

            "authority_expectation":
                "PRIMARY_AND_EXPLICIT_NONPRIMARY_DISTINCT",
        },


        "mixed_primary_signal_case": {
            "task_family":
                "MIXED_AUTHORITY_CONTEXT_RETRIEVAL",

            "evaluation_goal":
                (
                    "Retrieve evidence from files containing mixed primary "
                    "signals without flattening all utterances into one authority class."
                ),

            "evidence_scope":
                "PHYSICAL_FILE_MIXED_AUTHORITY",

            "required_dimensions": [
                "retrieval_hit",
                "attribution_correctness",
                "source_authority_correctness",
                "unnecessary_context",
            ],

            "context_expectation":
                "AUTHORITY_SIGNALS_REMAIN_LOCAL",

            "authority_expectation":
                "NO_FILE_LEVEL_AUTHORITY_FLATTENING",
        },


        "long_utterance_over_500": {
            "task_family":
                "LONG_UTTERANCE_RETRIEVAL",

            "evaluation_goal":
                (
                    "Retrieve sufficient evidence from a source utterance "
                    "longer than the 500-token candidate hard maximum."
                ),

            "evidence_scope":
                "SOURCE_UTTERANCE",

            "required_dimensions": [
                "retrieval_hit",
                "rank",
                "evidence_completeness",
                "citation_locator_correctness",
                "unnecessary_context",
            ],

            "context_expectation":
                "MULTI_FRAGMENT_IF_REQUIRED",

            "authority_expectation":
                "PRESERVE_SOURCE_ATTRIBUTION",
        },


        "long_utterance_over_800": {
            "task_family":
                "VERY_LONG_UTTERANCE_RETRIEVAL",

            "evaluation_goal":
                (
                    "Test evidence completeness for utterances exceeding "
                    "even the Medium candidate hard maximum."
                ),

            "evidence_scope":
                "SOURCE_UTTERANCE",

            "required_dimensions": [
                "retrieval_hit",
                "rank",
                "evidence_completeness",
                "citation_locator_correctness",
                "unnecessary_context",
            ],

            "context_expectation":
                "MULTI_FRAGMENT_REQUIRED_IF_NEEDED",

            "authority_expectation":
                "PRESERVE_SOURCE_ATTRIBUTION",
        },


        "token_fallback_sentence_case": {
            "task_family":
                "TOKEN_FALLBACK_FRAGMENT_RETRIEVAL",

            "evaluation_goal":
                (
                    "Test retrieval and evidence reconstruction when a single "
                    "source sentence itself exceeds the 500-token hard maximum."
                ),

            "evidence_scope":
                "SOURCE_SENTENCE_TOKEN_FALLBACK",

            "required_dimensions": [
                "retrieval_hit",
                "rank",
                "evidence_completeness",
                "citation_locator_correctness",
                "unnecessary_context",
            ],

            "context_expectation":
                "TOKEN_WINDOW_FRAGMENT_RECONSTRUCTION",

            "authority_expectation":
                "PRESERVE_SOURCE_ATTRIBUTION",
        },


        "question_answer_transition": {
            "task_family":
                "QUESTION_ANSWER_CONTEXT_CONFIRMATION",

            "evaluation_goal":
                (
                    "Confirm whether local transcript context actually contains "
                    "a question-answer relationship before constructing an "
                    "evaluation question."
                ),

            "evidence_scope":
                "ADJACENT_OR_LOCAL_TRANSCRIPT_CONTEXT",

            "required_dimensions": [
                "retrieval_hit",
                "evidence_completeness",
                "attribution_correctness",
                "unnecessary_context",
                "source_authority_correctness",
            ],

            "context_expectation":
                "QA_RELATION_MUST_BE_HUMAN_CONFIRMED",

            "authority_expectation":
                "NO_QA_OR_SPEAKER_INFERENCE",
        },
    },
}


# ============================================================
# 2) Validate exact 27-stratum coverage
# ============================================================

task_design_rows = []


for source_type in [
    "website",
    "book",
    "transcript",
]:

    expected = set(
        expected_strata[
            source_type
        ]
    )

    designed = set(
        TASK_DESIGN_BY_STRATUM[
            source_type
        ].keys()
    )

    assert (
        expected
        ==
        designed
    ), (
        f"Task design mismatch for {source_type}. "
        f"Missing={sorted(expected - designed)} "
        f"Extra={sorted(designed - expected)}"
    )


    for stratum in (
        expected_strata[
            source_type
        ]
    ):

        design = (
            TASK_DESIGN_BY_STRATUM[
                source_type
            ][
                stratum
            ]
        )

        case_records = [
            record
            for record
            in golden_case_records
            if (
                record[
                    "source_type"
                ]
                == source_type

                and

                record[
                    "stratum"
                ]
                == stratum
            )
        ]


        final_case_count = sum(
            1
            for record
            in case_records
            if (
                record[
                    "final_golden_membership"
                ]
                is True
            )
        )


        review_case_count = (
            len(
                case_records
            )
            -
            final_case_count
        )


        if (
            review_case_count
            > 0
        ):

            task_construction_status = (
                "BLOCKED_PENDING_HUMAN_CONFIRMATION"
            )

        else:

            task_construction_status = (
                "READY_FOR_TASK_CONSTRUCTION"
            )


        task_design_rows.append(
            {
                "source_type":
                    source_type,

                "stratum":
                    stratum,

                "task_family":
                    design[
                        "task_family"
                    ],

                "evaluation_goal":
                    design[
                        "evaluation_goal"
                    ],

                "evidence_scope":
                    design[
                        "evidence_scope"
                    ],

                "required_dimensions":
                    design[
                        "required_dimensions"
                    ],

                "context_expectation":
                    design[
                        "context_expectation"
                    ],

                "authority_expectation":
                    design[
                        "authority_expectation"
                    ],

                "selected_case_count":
                    len(
                        case_records
                    ),

                "final_case_count":
                    final_case_count,

                "review_pending_case_count":
                    review_case_count,

                "task_construction_status":
                    task_construction_status,
            }
        )


# ============================================================
# 3) Case-level task eligibility
# ============================================================

case_task_design = {}


for case_record in (
    golden_case_records
):

    source_type = (
        case_record[
            "source_type"
        ]
    )

    stratum = (
        case_record[
            "stratum"
        ]
    )

    design = (
        TASK_DESIGN_BY_STRATUM[
            source_type
        ][
            stratum
        ]
    )


    if (
        case_record[
            "final_golden_membership"
        ]
        is True
    ):

        construction_status = (
            "READY_FOR_TASK_CONSTRUCTION"
        )

    else:

        construction_status = (
            "BLOCKED_PENDING_HUMAN_CONFIRMATION"
        )


    case_task_design[
        case_record[
            "golden_case_id"
        ]
    ] = {
        "golden_case_id":
            case_record[
                "golden_case_id"
            ],

        "task_family":
            design[
                "task_family"
            ],

        "evidence_scope":
            design[
                "evidence_scope"
            ],

        "required_dimensions":
            list(
                design[
                    "required_dimensions"
                ]
            ),

        "context_expectation":
            design[
                "context_expectation"
            ],

        "authority_expectation":
            design[
                "authority_expectation"
            ],

        "task_construction_status":
            construction_status,

        "question_generation_status":
            "NOT_STARTED",

        "expected_answer_generation_status":
            "NOT_STARTED",
    }


# ============================================================
# 4) Task Design Contract
# ============================================================

EVALUATION_TASK_DESIGN_CONTRACT_V1_DRAFT = {
    "version":
        "evaluation_task_design_contract_v1_draft",

    "status":
        "VALIDATED_IN_MEMORY_NOT_FROZEN",

    "case_contract_version":
        GOLDEN_CASE_SCHEMA_VERSION,

    "frozen_evaluation_dimensions":
        FROZEN_EVALUATION_DIMENSIONS,

    "stratum_count":
        len(
            task_design_rows
        ),

    "unknown_speaker_inference":
        "FORBIDDEN",

    "authority_policy": {
        "direct_author":
            "PRIMARY",

        "supervised_website":
            "SECONDARY",

        "unresolved_transcript":
            "DO_NOT_INFER",

        "explicit_nonprimary":
            "KEEP_DISTINCT_FROM_PRIMARY",
    },

    "review_policy":
        (
            "NO_TASK_OR_QUESTION_CONSTRUCTION_FOR_REVIEW_CASES "
            "UNTIL HUMAN_CONFIRMATION"
        ),

    "question_generation_status":
        "NOT_STARTED",

    "expected_answer_generation_status":
        "NOT_STARTED",

    "retrieval_execution_status":
        "NOT_STARTED",

    "artifact_write_status":
        "NONE",

    "strata":
        task_design_rows,
}


# ============================================================
# 5) Coverage / integrity audits
# ============================================================

invalid_dimension_rows = []

empty_dimension_rows = []

review_task_leakage = []

case_design_missing = []


for row in (
    task_design_rows
):

    required_dimensions = (
        row[
            "required_dimensions"
        ]
    )

    if not required_dimensions:

        empty_dimension_rows.append(
            (
                row[
                    "source_type"
                ],
                row[
                    "stratum"
                ],
            )
        )


    invalid_dimensions = [
        dimension
        for dimension
        in required_dimensions
        if (
            dimension
            not in
            FROZEN_EVALUATION_DIMENSIONS
        )
    ]


    if invalid_dimensions:

        invalid_dimension_rows.append(
            {
                "source_type":
                    row[
                        "source_type"
                    ],

                "stratum":
                    row[
                        "stratum"
                    ],

                "invalid_dimensions":
                    invalid_dimensions,
            }
        )


for case_record in (
    golden_case_records
):

    case_id = (
        case_record[
            "golden_case_id"
        ]
    )

    if (
        case_id
        not in
        case_task_design
    ):

        case_design_missing.append(
            case_id
        )

        continue


    if (
        case_record[
            "final_golden_membership"
        ]
        is False

        and

        case_task_design[
            case_id
        ][
            "task_construction_status"
        ]
        !=
        "BLOCKED_PENDING_HUMAN_CONFIRMATION"
    ):

        review_task_leakage.append(
            case_id
        )


# ============================================================
# 6) Summary counts
# ============================================================

ready_case_count = sum(
    1
    for design
    in case_task_design.values()
    if (
        design[
            "task_construction_status"
        ]
        ==
        "READY_FOR_TASK_CONSTRUCTION"
    )
)


blocked_case_count = sum(
    1
    for design
    in case_task_design.values()
    if (
        design[
            "task_construction_status"
        ]
        ==
        "BLOCKED_PENDING_HUMAN_CONFIRMATION"
    )
)


ready_strata_count = sum(
    1
    for row
    in task_design_rows
    if (
        row[
            "task_construction_status"
        ]
        ==
        "READY_FOR_TASK_CONSTRUCTION"
    )
)


blocked_strata_count = sum(
    1
    for row
    in task_design_rows
    if (
        row[
            "task_construction_status"
        ]
        ==
        "BLOCKED_PENDING_HUMAN_CONFIRMATION"
    )
)


dimension_usage_counts = Counter(
    dimension

    for row
    in task_design_rows

    for dimension
    in row[
        "required_dimensions"
    ]
)


# ============================================================
# 7) Validation gates
# ============================================================

exact_27_task_designs_gate = (
    len(
        task_design_rows
    )
    == 27
)


all_dimensions_valid_gate = (
    len(
        invalid_dimension_rows
    )
    == 0
)


no_empty_dimensions_gate = (
    len(
        empty_dimension_rows
    )
    == 0
)


all_724_cases_have_design_gate = (
    len(
        case_task_design
    )
    ==
    len(
        golden_case_records
    )
    ==
    724

    and

    len(
        case_design_missing
    )
    == 0
)


exact_572_ready_cases_gate = (
    ready_case_count
    == 572
)


exact_152_blocked_cases_gate = (
    blocked_case_count
    == 152
)


exact_23_ready_strata_gate = (
    ready_strata_count
    == 23
)


exact_4_blocked_strata_gate = (
    blocked_strata_count
    == 4
)


no_review_task_leakage_gate = (
    len(
        review_task_leakage
    )
    == 0
)


all_frozen_dimensions_used_gate = (
    set(
        dimension_usage_counts
    )
    ==
    set(
        FROZEN_EVALUATION_DIMENSIONS
    )
)


unknown_speaker_guardrail_gate = (
    EVALUATION_TASK_DESIGN_CONTRACT_V1_DRAFT[
        "unknown_speaker_inference"
    ]
    ==
    "FORBIDDEN"
)


question_generation_not_started_gate = all(
    design[
        "question_generation_status"
    ]
    == "NOT_STARTED"

    and

    design[
        "expected_answer_generation_status"
    ]
    == "NOT_STARTED"

    for design
    in case_task_design.values()
)


retrieval_not_started_gate = (
    EVALUATION_TASK_DESIGN_CONTRACT_V1_DRAFT[
        "retrieval_execution_status"
    ]
    == "NOT_STARTED"
)


no_artifact_write_gate = (
    EVALUATION_TASK_DESIGN_CONTRACT_V1_DRAFT[
        "artifact_write_status"
    ]
    == "NONE"
)


cell_24_gates = {
    "exact_27_stratum_task_designs":
        exact_27_task_designs_gate,

    "all_required_dimensions_from_frozen_contract":
        all_dimensions_valid_gate,

    "no_stratum_has_empty_evaluation_dimensions":
        no_empty_dimensions_gate,

    "all_724_cases_have_task_design":
        all_724_cases_have_design_gate,

    "exact_572_cases_ready_for_task_construction":
        exact_572_ready_cases_gate,

    "exact_152_cases_blocked_pending_review":
        exact_152_blocked_cases_gate,

    "exact_23_strata_ready_for_task_construction":
        exact_23_ready_strata_gate,

    "exact_4_strata_blocked_pending_review":
        exact_4_blocked_strata_gate,

    "no_review_case_leaks_into_task_construction":
        no_review_task_leakage_gate,

    "all_7_frozen_evaluation_dimensions_used":
        all_frozen_dimensions_used_gate,

    "unknown_speaker_inference_forbidden":
        unknown_speaker_guardrail_gate,

    "question_and_answer_generation_not_started":
        question_generation_not_started_gate,

    "retrieval_execution_not_started":
        retrieval_not_started_gate,

    "no_artifact_write":
        no_artifact_write_gate,
}


cell_24_pass = all(
    cell_24_gates.values()
)


# ============================================================
# 8) Output
# ============================================================

print(
    "=== EVALUATION TASK DESIGN CONTRACT ==="
)

print(
    "Stratum task designs:",
    len(
        task_design_rows
    )
)

print(
    "Ready strata:",
    ready_strata_count
)

print(
    "Blocked review strata:",
    blocked_strata_count
)

print(
    "Ready cases:",
    ready_case_count
)

print(
    "Blocked review cases:",
    blocked_case_count
)


print(
    "\n=== DIMENSION USAGE ==="
)

for dimension in (
    FROZEN_EVALUATION_DIMENSIONS
):

    print(
        "-",
        dimension,
        ":",
        dimension_usage_counts[
            dimension
        ],
        "strata",
    )


print(
    "\n=== PER-STRATUM TASK DESIGN ==="
)

for row in (
    task_design_rows
):

    print(
        "-",
        row[
            "source_type"
        ],
        "/",
        row[
            "stratum"
        ],
        "| family:",
        row[
            "task_family"
        ],
        "| evidence:",
        row[
            "evidence_scope"
        ],
        "| selected:",
        row[
            "selected_case_count"
        ],
        "| final:",
        row[
            "final_case_count"
        ],
        "| review:",
        row[
            "review_pending_case_count"
        ],
        "| status:",
        row[
            "task_construction_status"
        ],
    )

    print(
        "   dimensions:",
        row[
            "required_dimensions"
        ]
    )


print(
    "\n=== BLOCKED REVIEW STRATA ==="
)

for row in (
    task_design_rows
):

    if (
        row[
            "task_construction_status"
        ]
        ==
        "BLOCKED_PENDING_HUMAN_CONFIRMATION"
    ):

        print(
            "-",
            row[
                "source_type"
            ],
            "/",
            row[
                "stratum"
            ],
            "| cases:",
            row[
                "review_pending_case_count"
            ],
        )


print(
    "\n=== CODE CELL 24 GATES ==="
)

for name, passed in (
    cell_24_gates.items()
):

    print(
        name,
        ":",
        "PASS"
        if passed
        else "FAIL"
    )


print(
    "\nCODE CELL 24 GATE:",
    "PASS"
    if cell_24_pass
    else "FAIL"
)


assert cell_24_pass, (
    "Evaluation Task Design Contract validation failed. "
    "Do not generate questions or expected answers until every "
    "Golden Case has a valid task family, evidence scope, authority "
    "expectation, and frozen evaluation-dimension mapping."
)
```

<div style="direction: rtl; white-space: normal; line-height: 1.7;">

<h3>Question Construction Guardrails and Groundability Contract</h3>

<p>
Task Design برای تمام Caseها مشخص شده است، اما هنوز هیچ سؤال Golden نباید تولید شود
تا Groundability و محدودیت‌های Question Construction به‌صورت صریح تعریف و Validate شوند.
</p>

<p>
اصل اصلی این مرحله این است:
هر سؤال آینده باید فقط از Evidence واقعی همان Golden Case قابل پاسخ باشد
و نباید برای پاسخ صحیح به دانش خارجی، حدس، Speaker Inference،
Authority Promotion یا Semantic Assumption نیاز داشته باشد.
</p>

<h4>قواعد اصلی</h4>

<ul>
<li>Question باید از Evidence بازیابی‌شده‌ی همان Case قابل Ground باشد.</li>
<li>Evidence Text باید واقعاً در Accepted Corpus موجود و غیرخالی باشد.</li>
<li>Direct Author، Supervised Secondary، Contextual و Unresolved Authority نباید با هم ادغام شوند.</li>
<li>Unknown/Unresolved Speaker هرگز نباید حدس زده شود.</li>
<li>وجود Similarity Relation به معنی Duplicate یا Citation Correctness قطعی نیست.</li>
<li>Question/Answer Transition فقط پس از Human Confirmation مجاز است.</li>
<li>هیچ سؤال نباید Permission یا Release Right را از Processing Rights استنتاج کند.</li>
<li>هیچ سؤال نباید برای پاسخ صحیح نیازمند External Knowledge باشد.</li>
<li>اگر Evidence برای پاسخ کافی نباشد، Case باید به <code>ABSTAIN_OR_REVIEW</code> هدایت شود، نه اینکه Answer ساخته شود.</li>
</ul>

<p>
این Cell فقط Guardrail و Groundability Precheck را اجرا می‌کند.
هیچ Question، Expected Answer، Citation Target یا Retrieval Judgment تولید نمی‌شود.
</p>

</div>

```python
# Code Cell 25

# ============================================================
# Question Construction Guardrails + Groundability Precheck
#
# NO question generation.
# NO expected-answer generation.
# NO retrieval execution.
# NO artifact writes.
# ============================================================


# ============================================================
# 1) Build a lightweight accepted-corpus text index
# ============================================================

corpus_text_index = {}

empty_text_document_ids = []


with corpus_path.open(
    "r",
    encoding="utf-8",
) as handle:

    for line in handle:

        if not line.strip():
            continue

        record = json.loads(
            line
        )

        document_id = (
            record.get(
                "document_id"
            )
        )

        text = str(
            record.get(
                "text"
            )
            or ""
        )

        corpus_text_index[
            document_id
        ] = text

        if not text.strip():

            empty_text_document_ids.append(
                document_id
            )


# ============================================================
# 2) Frozen Question Construction Guardrails
# ============================================================

QUESTION_CONSTRUCTION_GUARDRAILS_V1_DRAFT = {
    "version":
        "question_construction_guardrails_v1_draft",

    "status":
        "VALIDATED_IN_MEMORY_NOT_FROZEN",

    "grounding_policy": {
        "answer_must_be_supported_by_case_evidence":
            True,

        "external_knowledge_required":
            False,

        "unsupported_fact_completion":
            "FORBIDDEN",

        "fabricated_evidence":
            "FORBIDDEN",

        "fabricated_locator":
            "FORBIDDEN",
    },

    "attribution_policy": {
        "unknown_speaker_inference":
            "FORBIDDEN",

        "unresolved_speaker_promotion":
            "FORBIDDEN",

        "secondary_to_primary_promotion":
            "FORBIDDEN",

        "file_level_authority_flattening":
            "FORBIDDEN",

        "multi_speaker_collapse":
            "FORBIDDEN",
    },

    "similarity_policy": {
        "similarity_means_duplicate":
            False,

        "similarity_means_same_authority":
            False,

        "similarity_means_same_locator":
            False,
    },

    "rights_policy": {
        "processing_allowed_implies_release_allowed":
            False,

        "permission_inference":
            "FORBIDDEN",
    },

    "qa_policy": {
        "speaker_transition_proves_question_answer":
            False,

        "qa_relation_requires_explicit_or_human_confirmation":
            True,
    },

    "failure_policy": {
        "insufficient_evidence":
            "ABSTAIN_OR_REVIEW",

        "ambiguous_attribution":
            "ABSTAIN_OR_REVIEW",

        "ambiguous_locator":
            "ABSTAIN_OR_REVIEW",

        "review_case_without_confirmation":
            "BLOCKED",
    },

    "question_generation_status":
        "NOT_STARTED",

    "expected_answer_generation_status":
        "NOT_STARTED",

    "artifact_write_status":
        "NONE",
}


# ============================================================
# 3) Stratum-specific construction constraints
# ============================================================

STRATUM_QUESTION_CONSTRAINTS_V1_DRAFT = {
    "website": {
        "direct_author_short": {
            "allow_direct_dr_roudi_authority_claim":
                True,

            "require_local_source_grounding":
                True,
        },

        "direct_author_long": {
            "allow_direct_dr_roudi_authority_claim":
                True,

            "require_local_source_grounding":
                True,

            "multi_part_evidence_allowed":
                True,
        },

        "supervised_short": {
            "allow_direct_dr_roudi_authority_claim":
                False,

            "preserve_supervised_secondary_status":
                True,
        },

        "supervised_long": {
            "allow_direct_dr_roudi_authority_claim":
                False,

            "preserve_supervised_secondary_status":
                True,

            "multi_part_evidence_allowed":
                True,
        },

        "multi_topic_article": {
            "requires_human_confirmation_before_question":
                True,
        },

        "citation_locator_sensitive_case": {
            "requires_human_confirmation_before_question":
                True,

            "semantic_similarity_not_sufficient":
                True,
        },
    },

    "book": {
        "intrinsically_tiny_narrative_region": {
            "preserve_region_boundary":
                True,
        },

        "intrinsically_tiny_table_region": {
            "preserve_table_boundary":
                True,
        },

        "long_section": {
            "multi_document_evidence_allowed":
                True,
        },

        "large_table_run": {
            "preserve_table_run":
                True,

            "multi_document_evidence_allowed":
                True,
        },

        "exercise_or_self_assessment": {
            "preserve_instructional_function":
                True,
        },

        "supplemental_footnote_case": {
            "parent_context_may_be_required":
                True,
        },

        "deep_heading_path": {
            "preserve_heading_path_for_citation":
                True,
        },

        "multi_paragraph_evidence_case": {
            "requires_human_confirmation_before_question":
                True,
        },
    },

    "transcript": {
        "single_speaker_primary": {
            "primary_attribution_allowed_only_from_recorded_evidence":
                True,
        },

        "single_speaker_unresolved": {
            "speaker_identity_question_requiring_inference":
                "FORBIDDEN",

            "unresolved_identity_must_remain_unresolved":
                True,
        },

        "multi_speaker_uni_tehran": {
            "utterance_level_attribution_required":
                True,

            "file_level_authority_flattening":
                "FORBIDDEN",
        },

        "multi_speaker_chehelston": {
            "utterance_level_attribution_required":
                True,

            "file_level_authority_flattening":
                "FORBIDDEN",
        },

        "multi_speaker_asre_shirin": {
            "explicit_label_is_not_primary_authority_proof":
                True,
        },

        "multi_speaker_khane_to": {
            "speaker_boundary_preservation_required":
                True,
        },

        "multi_speaker_mizan": {
            "unknown_speaker_inference":
                "FORBIDDEN",
        },

        "primary_to_nonprimary_transition": {
            "left_right_authority_must_remain_distinct":
                True,
        },

        "mixed_primary_signal_case": {
            "file_level_authority_flattening":
                "FORBIDDEN",
        },

        "long_utterance_over_500": {
            "multi_fragment_evidence_allowed":
                True,
        },

        "long_utterance_over_800": {
            "multi_fragment_evidence_allowed":
                True,
        },

        "token_fallback_sentence_case": {
            "token_window_reconstruction_may_be_required":
                True,
        },

        "question_answer_transition": {
            "requires_human_confirmation_before_question":
                True,

            "speaker_transition_alone_is_not_qa_proof":
                True,
        },
    },
}


# ============================================================
# 4) Validate exact stratum coverage
# ============================================================

constraint_coverage_failures = []


for source_type in [
    "website",
    "book",
    "transcript",
]:

    expected = set(
        expected_strata[
            source_type
        ]
    )

    actual = set(
        STRATUM_QUESTION_CONSTRAINTS_V1_DRAFT[
            source_type
        ].keys()
    )

    if expected != actual:

        constraint_coverage_failures.append(
            {
                "source_type":
                    source_type,

                "missing":
                    sorted(
                        expected - actual
                    ),

                "extra":
                    sorted(
                        actual - expected
                    ),
            }
        )


# ============================================================
# 5) Case-level Groundability Precheck
# ============================================================

case_groundability_precheck = {}

missing_evidence_text_cases = []

empty_case_evidence_text_cases = []

review_case_readiness_leakage = []

task_status_mismatches = []


for case_record in (
    golden_case_records
):

    case_id = (
        case_record[
            "golden_case_id"
        ]
    )

    source_type = (
        case_record[
            "source_type"
        ]
    )

    stratum = (
        case_record[
            "stratum"
        ]
    )

    evidence_document_ids = (
        case_record[
            "evidence_document_ids"
        ]
    )

    task_design = (
        case_task_design[
            case_id
        ]
    )


    missing_document_ids = [
        document_id
        for document_id
        in evidence_document_ids
        if (
            document_id
            not in corpus_text_index
        )
    ]


    empty_document_ids = [
        document_id
        for document_id
        in evidence_document_ids
        if (
            document_id
            in corpus_text_index

            and

            not corpus_text_index[
                document_id
            ].strip()
        )
    ]


    evidence_text_recoverable = (
        len(
            evidence_document_ids
        )
        > 0

        and

        len(
            missing_document_ids
        )
        == 0

        and

        len(
            empty_document_ids
        )
        == 0
    )


    if missing_document_ids:

        missing_evidence_text_cases.append(
            {
                "golden_case_id":
                    case_id,

                "missing_document_ids":
                    missing_document_ids,
            }
        )


    if empty_document_ids:

        empty_case_evidence_text_cases.append(
            {
                "golden_case_id":
                    case_id,

                "empty_document_ids":
                    empty_document_ids,
            }
        )


    # --------------------------------------------------------
    # Review cases remain blocked regardless of evidence text.
    # --------------------------------------------------------

    if (
        case_record[
            "final_golden_membership"
        ]
        is False
    ):

        construction_eligibility = (
            "BLOCKED_PENDING_HUMAN_CONFIRMATION"
        )

        groundability_precheck_status = (
            "REVIEW_REQUIRED_BEFORE_GROUNDABILITY_DECISION"
        )

    else:

        if evidence_text_recoverable:

            construction_eligibility = (
                "READY_FOR_GROUNDED_TASK_DRAFTING"
            )

            groundability_precheck_status = (
                "EVIDENCE_RECOVERABILITY_PASS"
            )

        else:

            construction_eligibility = (
                "ABSTAIN_OR_REVIEW"
            )

            groundability_precheck_status = (
                "EVIDENCE_RECOVERABILITY_FAIL"
            )


    if (
        case_record[
            "final_golden_membership"
        ]
        is False

        and

        construction_eligibility
        !=
        "BLOCKED_PENDING_HUMAN_CONFIRMATION"
    ):

        review_case_readiness_leakage.append(
            case_id
        )


    expected_task_status = (
        "READY_FOR_TASK_CONSTRUCTION"
        if (
            case_record[
                "final_golden_membership"
            ]
            is True
        )
        else
        "BLOCKED_PENDING_HUMAN_CONFIRMATION"
    )


    if (
        task_design[
            "task_construction_status"
        ]
        != expected_task_status
    ):

        task_status_mismatches.append(
            case_id
        )


    case_groundability_precheck[
        case_id
    ] = {
        "golden_case_id":
            case_id,

        "source_type":
            source_type,

        "stratum":
            stratum,

        "evidence_document_count":
            len(
                evidence_document_ids
            ),

        "evidence_text_recoverable":
            evidence_text_recoverable,

        "groundability_precheck_status":
            groundability_precheck_status,

        "construction_eligibility":
            construction_eligibility,

        "question_generation_status":
            "NOT_STARTED",

        "expected_answer_generation_status":
            "NOT_STARTED",

        "external_knowledge_required":
            False,

        "unknown_speaker_inference":
            "FORBIDDEN",
    }


# ============================================================
# 6) Summary counts
# ============================================================

ready_for_grounded_drafting_count = sum(
    1
    for item
    in case_groundability_precheck.values()
    if (
        item[
            "construction_eligibility"
        ]
        ==
        "READY_FOR_GROUNDED_TASK_DRAFTING"
    )
)


blocked_review_count = sum(
    1
    for item
    in case_groundability_precheck.values()
    if (
        item[
            "construction_eligibility"
        ]
        ==
        "BLOCKED_PENDING_HUMAN_CONFIRMATION"
    )
)


abstain_or_review_count = sum(
    1
    for item
    in case_groundability_precheck.values()
    if (
        item[
            "construction_eligibility"
        ]
        ==
        "ABSTAIN_OR_REVIEW"
    )
)


# ============================================================
# 7) Groundability Contract
# ============================================================

GROUNDABILITY_CONTRACT_V1_DRAFT = {
    "version":
        "groundability_contract_v1_draft",

    "status":
        "VALIDATED_IN_MEMORY_NOT_FROZEN",

    "case_count":
        len(
            case_groundability_precheck
        ),

    "ready_for_grounded_task_drafting":
        ready_for_grounded_drafting_count,

    "blocked_pending_human_confirmation":
        blocked_review_count,

    "abstain_or_review":
        abstain_or_review_count,

    "groundability_definition": (
        "A future evaluation task is groundable only when its "
        "correct resolution can be supported by the case evidence "
        "without external knowledge, unsupported attribution, "
        "speaker inference, locator fabrication, or authority promotion."
    ),

    "important_limitation": (
        "EVIDENCE_RECOVERABILITY_PASS does not prove that a useful "
        "question exists; semantic question groundability must be "
        "validated after a task draft is constructed."
    ),

    "question_generation_status":
        "NOT_STARTED",

    "expected_answer_generation_status":
        "NOT_STARTED",

    "artifact_write_status":
        "NONE",
}


# ============================================================
# 8) Validation gates
# ============================================================

exact_stratum_constraint_coverage_gate = (
    len(
        constraint_coverage_failures
    )
    == 0
)


all_724_cases_prechecked_gate = (
    len(
        case_groundability_precheck
    )
    ==
    len(
        golden_case_records
    )
    ==
    724
)


all_case_evidence_text_recoverable_gate = (
    len(
        missing_evidence_text_cases
    )
    == 0

    and

    len(
        empty_case_evidence_text_cases
    )
    == 0
)


exact_572_ready_for_grounded_drafting_gate = (
    ready_for_grounded_drafting_count
    == 572
)


exact_152_review_blocked_gate = (
    blocked_review_count
    == 152
)


zero_unexpected_abstain_gate = (
    abstain_or_review_count
    == 0
)


no_review_readiness_leakage_gate = (
    len(
        review_case_readiness_leakage
    )
    == 0
)


task_status_alignment_gate = (
    len(
        task_status_mismatches
    )
    == 0
)


unknown_speaker_guardrail_gate = (
    QUESTION_CONSTRUCTION_GUARDRAILS_V1_DRAFT[
        "attribution_policy"
    ][
        "unknown_speaker_inference"
    ]
    ==
    "FORBIDDEN"

    and

    all(
        item[
            "unknown_speaker_inference"
        ]
        ==
        "FORBIDDEN"

        for item
        in case_groundability_precheck.values()
    )
)


no_external_knowledge_gate = (
    QUESTION_CONSTRUCTION_GUARDRAILS_V1_DRAFT[
        "grounding_policy"
    ][
        "external_knowledge_required"
    ]
    is False

    and

    all(
        item[
            "external_knowledge_required"
        ]
        is False

        for item
        in case_groundability_precheck.values()
    )
)


similarity_not_duplicate_gate = (
    QUESTION_CONSTRUCTION_GUARDRAILS_V1_DRAFT[
        "similarity_policy"
    ][
        "similarity_means_duplicate"
    ]
    is False
)


rights_not_inferred_gate = (
    QUESTION_CONSTRUCTION_GUARDRAILS_V1_DRAFT[
        "rights_policy"
    ][
        "processing_allowed_implies_release_allowed"
    ]
    is False

    and

    QUESTION_CONSTRUCTION_GUARDRAILS_V1_DRAFT[
        "rights_policy"
    ][
        "permission_inference"
    ]
    ==
    "FORBIDDEN"
)


qa_inference_forbidden_gate = (
    QUESTION_CONSTRUCTION_GUARDRAILS_V1_DRAFT[
        "qa_policy"
    ][
        "speaker_transition_proves_question_answer"
    ]
    is False
)


question_generation_not_started_gate = (
    QUESTION_CONSTRUCTION_GUARDRAILS_V1_DRAFT[
        "question_generation_status"
    ]
    == "NOT_STARTED"

    and

    GROUNDABILITY_CONTRACT_V1_DRAFT[
        "question_generation_status"
    ]
    == "NOT_STARTED"

    and

    all(
        item[
            "question_generation_status"
        ]
        == "NOT_STARTED"

        for item
        in case_groundability_precheck.values()
    )
)


expected_answer_not_started_gate = (
    GROUNDABILITY_CONTRACT_V1_DRAFT[
        "expected_answer_generation_status"
    ]
    == "NOT_STARTED"

    and

    all(
        item[
            "expected_answer_generation_status"
        ]
        == "NOT_STARTED"

        for item
        in case_groundability_precheck.values()
    )
)


no_artifact_write_gate = (
    QUESTION_CONSTRUCTION_GUARDRAILS_V1_DRAFT[
        "artifact_write_status"
    ]
    == "NONE"

    and

    GROUNDABILITY_CONTRACT_V1_DRAFT[
        "artifact_write_status"
    ]
    == "NONE"
)


cell_25_gates = {
    "exact_27_stratum_constraint_coverage":
        exact_stratum_constraint_coverage_gate,

    "all_724_cases_groundability_prechecked":
        all_724_cases_prechecked_gate,

    "all_selected_case_evidence_text_recoverable":
        all_case_evidence_text_recoverable_gate,

    "exact_572_ready_for_grounded_task_drafting":
        exact_572_ready_for_grounded_drafting_gate,

    "exact_152_review_cases_remain_blocked":
        exact_152_review_blocked_gate,

    "zero_unexpected_abstain_or_review_cases":
        zero_unexpected_abstain_gate,

    "no_review_case_leaks_into_drafting":
        no_review_readiness_leakage_gate,

    "task_design_and_groundability_status_aligned":
        task_status_alignment_gate,

    "unknown_speaker_inference_forbidden":
        unknown_speaker_guardrail_gate,

    "external_knowledge_not_required":
        no_external_knowledge_gate,

    "similarity_not_treated_as_duplicate":
        similarity_not_duplicate_gate,

    "rights_or_permission_not_inferred":
        rights_not_inferred_gate,

    "speaker_transition_not_treated_as_qa_proof":
        qa_inference_forbidden_gate,

    "question_generation_not_started":
        question_generation_not_started_gate,

    "expected_answer_generation_not_started":
        expected_answer_not_started_gate,

    "no_artifact_write":
        no_artifact_write_gate,
}


cell_25_pass = all(
    cell_25_gates.values()
)


# ============================================================
# 9) Output
# ============================================================

print(
    "=== QUESTION CONSTRUCTION & GROUNDABILITY CONTRACT ==="
)

print(
    "Cases prechecked:",
    len(
        case_groundability_precheck
    )
)

print(
    "Ready for grounded task drafting:",
    ready_for_grounded_drafting_count
)

print(
    "Blocked pending human confirmation:",
    blocked_review_count
)

print(
    "Abstain or review:",
    abstain_or_review_count
)


print(
    "\n=== CORPUS TEXT RECOVERABILITY ==="
)

print(
    "Corpus documents indexed:",
    len(
        corpus_text_index
    )
)

print(
    "Empty corpus documents:",
    len(
        empty_text_document_ids
    )
)

print(
    "Cases with missing evidence text:",
    len(
        missing_evidence_text_cases
    )
)

print(
    "Cases with empty evidence text:",
    len(
        empty_case_evidence_text_cases
    )
)


print(
    "\n=== CORE GUARDRAILS ==="
)

print(
    "- Unknown speaker inference:",
    QUESTION_CONSTRUCTION_GUARDRAILS_V1_DRAFT[
        "attribution_policy"
    ][
        "unknown_speaker_inference"
    ]
)

print(
    "- Secondary -> Primary promotion:",
    QUESTION_CONSTRUCTION_GUARDRAILS_V1_DRAFT[
        "attribution_policy"
    ][
        "secondary_to_primary_promotion"
    ]
)

print(
    "- Similarity means duplicate:",
    QUESTION_CONSTRUCTION_GUARDRAILS_V1_DRAFT[
        "similarity_policy"
    ][
        "similarity_means_duplicate"
    ]
)

print(
    "- Processing allowed implies release allowed:",
    QUESTION_CONSTRUCTION_GUARDRAILS_V1_DRAFT[
        "rights_policy"
    ][
        "processing_allowed_implies_release_allowed"
    ]
)

print(
    "- Speaker transition proves Q/A:",
    QUESTION_CONSTRUCTION_GUARDRAILS_V1_DRAFT[
        "qa_policy"
    ][
        "speaker_transition_proves_question_answer"
    ]
)


print(
    "\n=== CONSTRUCTION ELIGIBILITY COUNTS ==="
)

eligibility_counts = Counter(
    item[
        "construction_eligibility"
    ]
    for item
    in case_groundability_precheck.values()
)

for status, count in sorted(
    eligibility_counts.items()
):

    print(
        "-",
        status,
        ":",
        count,
    )


print(
    "\n=== CODE CELL 25 GATES ==="
)

for name, passed in (
    cell_25_gates.items()
):

    print(
        name,
        ":",
        "PASS"
        if passed
        else "FAIL"
    )


print(
    "\nCODE CELL 25 GATE:",
    "PASS"
    if cell_25_pass
    else "FAIL"
)


assert cell_25_pass, (
    "Question-construction groundability precheck failed. "
    "Do not draft Golden evaluation questions until evidence "
    "recoverability and attribution/authority/rights guardrails pass."
)
```

<div style="direction: rtl; white-space: normal; line-height: 1.7;">

<h3>Task Draft Schema and Question Construction Method</h3>

<p>
پس از Groundability Precheck، ۵۷۲ Golden Case برای Drafting آماده‌اند؛
اما Evidence Recoverability به‌تنهایی برای تولید سؤال کافی نیست.
قبل از نوشتن متن سؤال باید ساختار Task Draft و روش انتخاب Evidence Target مشخص شود.
</p>

<p>
هر Task Draft آینده باید سه لایه را از هم جدا نگه دارد:
</p>

<ul>
<li><strong>Case Evidence</strong>: تمام Evidence معتبر متصل به Golden Case.</li>
<li><strong>Target Evidence</strong>: بخش یا رکورد مشخصی که سؤال واقعاً درباره‌ی آن ساخته می‌شود.</li>
<li><strong>Question Text</strong>: سؤال طبیعی که فقط پس از تعیین و Validate شدن Target Evidence تولید خواهد شد.</li>
</ul>

<p>
این تفکیک مخصوصاً برای Caseهای بزرگ مانند Physical File، Book Section،
Long Website Document و Long Utterance ضروری است؛
زیرا کل Case Evidence نباید به‌صورت مبهم به‌عنوان Answer Target تلقی شود.
</p>

<h4>قواعد Question Construction</h4>

<ul>
<li>Question نباید از Document ID، Hash یا Metadata داخلی به‌عنوان Shortcut استفاده کند.</li>
<li>Question باید طبیعی و مستقل از نام‌گذاری داخلی Pipeline باشد.</li>
<li>Metadata فقط زمانی می‌تواند موضوع سؤال باشد که Task Family واقعاً Attribution، Authority یا Locator را ارزیابی کند.</li>
<li>Target Evidence باید قبل از Question Text تعیین شود.</li>
<li>Caseهای Review-pending هیچ Target یا Question قابل اجرا دریافت نمی‌کنند.</li>
<li>Unknown Speaker، Authority Promotion، Duplicate Inference و Rights Inference ممنوع باقی می‌مانند.</li>
<li>اگر Target Evidence معتبر قابل تعیین نباشد، Case باید به Review/Abstention برود؛ سؤال مصنوعی ساخته نمی‌شود.</li>
</ul>

<p>
این Cell فقط Schema، Archetype و Construction Method را ثبت می‌کند.
هیچ Evidence Target، Question Text، Expected Answer یا Citation Judgment هنوز تولید نمی‌شود.
</p>

</div>

```python
# Code Cell 26

# ============================================================
# Task Draft Schema + Question Construction Method
#
# DESIGN ONLY.
#
# NO target evidence selection.
# NO question text generation.
# NO expected-answer generation.
# NO retrieval execution.
# NO artifact writes.
# ============================================================


# ============================================================
# 1) Frozen question archetypes
# ============================================================

QUESTION_ARCHETYPES_V1 = {

    "CONTENT_RETRIEVAL":
        (
            "Ask a natural content question whose answer is directly "
            "grounded in the selected evidence target."
        ),

    "AUTHORITY_CONSTRAINED_RETRIEVAL":
        (
            "Ask about content while requiring the system to preserve "
            "the source's actual authority or attribution class."
        ),

    "EVIDENCE_COMPLETENESS":
        (
            "Ask a question whose correct support may require multiple "
            "parts of the same valid evidence scope."
        ),

    "STRUCTURE_AWARE_RETRIEVAL":
        (
            "Ask about evidence whose structural boundary, hierarchy, "
            "table context, footnote relation, or locator is material."
        ),

    "ATTRIBUTION_BOUNDARY":
        (
            "Ask about content near speaker or authority boundaries "
            "without collapsing distinct speakers or authority states."
        ),

    "UNRESOLVED_IDENTITY_ABSTENTION":
        (
            "Construct a task where unresolved identity must remain "
            "unresolved rather than being inferred."
        ),

    "FRAGMENT_RECONSTRUCTION":
        (
            "Ask about evidence that may span deterministic fragments "
            "created by the frozen chunking fallback logic."
        ),

    "REVIEW_CONFIRMATION_ONLY":
        (
            "No evaluation question may be constructed until the "
            "candidate's semantic membership is human-confirmed."
        ),
}


# ============================================================
# 2) Explicit archetype mapping for all 27 strata
#
# This is a design mapping only; it does not generate questions.
# ============================================================

QUESTION_ARCHETYPE_BY_STRATUM = {

    "website": {
        "direct_author_short":
            "AUTHORITY_CONSTRAINED_RETRIEVAL",

        "direct_author_long":
            "EVIDENCE_COMPLETENESS",

        "supervised_short":
            "AUTHORITY_CONSTRAINED_RETRIEVAL",

        "supervised_long":
            "EVIDENCE_COMPLETENESS",

        "multi_topic_article":
            "REVIEW_CONFIRMATION_ONLY",

        "citation_locator_sensitive_case":
            "REVIEW_CONFIRMATION_ONLY",
    },


    "book": {
        "intrinsically_tiny_narrative_region":
            "STRUCTURE_AWARE_RETRIEVAL",

        "intrinsically_tiny_table_region":
            "STRUCTURE_AWARE_RETRIEVAL",

        "long_section":
            "EVIDENCE_COMPLETENESS",

        "large_table_run":
            "STRUCTURE_AWARE_RETRIEVAL",

        "exercise_or_self_assessment":
            "STRUCTURE_AWARE_RETRIEVAL",

        "supplemental_footnote_case":
            "STRUCTURE_AWARE_RETRIEVAL",

        "deep_heading_path":
            "STRUCTURE_AWARE_RETRIEVAL",

        "multi_paragraph_evidence_case":
            "REVIEW_CONFIRMATION_ONLY",
    },


    "transcript": {
        "single_speaker_primary":
            "AUTHORITY_CONSTRAINED_RETRIEVAL",

        "single_speaker_unresolved":
            "UNRESOLVED_IDENTITY_ABSTENTION",

        "multi_speaker_uni_tehran":
            "ATTRIBUTION_BOUNDARY",

        "multi_speaker_chehelston":
            "ATTRIBUTION_BOUNDARY",

        "multi_speaker_asre_shirin":
            "ATTRIBUTION_BOUNDARY",

        "multi_speaker_khane_to":
            "ATTRIBUTION_BOUNDARY",

        "multi_speaker_mizan":
            "ATTRIBUTION_BOUNDARY",

        "primary_to_nonprimary_transition":
            "ATTRIBUTION_BOUNDARY",

        "mixed_primary_signal_case":
            "ATTRIBUTION_BOUNDARY",

        "long_utterance_over_500":
            "EVIDENCE_COMPLETENESS",

        "long_utterance_over_800":
            "EVIDENCE_COMPLETENESS",

        "token_fallback_sentence_case":
            "FRAGMENT_RECONSTRUCTION",

        "question_answer_transition":
            "REVIEW_CONFIRMATION_ONLY",
    },
}


# ============================================================
# 3) Evidence-target selection methods
#
# These define HOW the next stage must identify a precise
# evidence target before any question is drafted.
# ============================================================

EVIDENCE_TARGET_METHOD_BY_SCOPE = {

    "SINGLE_SOURCE_DOCUMENT":
        "DOCUMENT_LEVEL_TARGET",

    "SINGLE_LONG_SOURCE_DOCUMENT":
        "DETERMINISTIC_LOCAL_EVIDENCE_TARGET_REQUIRED",

    "LOCAL_EVIDENCE_WITHIN_SOURCE_DOCUMENT":
        "HUMAN_CONFIRMED_LOCAL_TARGET_REQUIRED",

    "SOURCE_DOCUMENT_PLUS_SIMILARITY_CONTRAST":
        "HUMAN_CONFIRMED_SOURCE_AND_CONTRAST_TARGET_REQUIRED",

    "PROTECTED_STRUCTURAL_REGION":
        "USE_EXACT_PROTECTED_REGION",

    "PROTECTED_TABLE_REGION":
        "USE_EXACT_PROTECTED_TABLE_REGION",

    "BOOK_SECTION":
        "DETERMINISTIC_LOCAL_OR_MULTI_RECORD_TARGET_REQUIRED",

    "CONTIGUOUS_TABLE_RUN":
        "USE_EXACT_TABLE_RUN",

    "BOOK_REGION":
        "USE_EXPLICIT_BOOK_REGION",

    "PARENT_PLUS_SUPPLEMENTAL_RECORD":
        "USE_PARENT_PLUS_SUPPLEMENTAL_RELATION",

    "BOOK_RECORD_WITH_HIERARCHICAL_CONTEXT":
        "USE_RECORD_PLUS_HEADING_PATH",

    "BOOK_SECTION_MULTI_PARAGRAPH":
        "HUMAN_CONFIRMED_MULTI_PARAGRAPH_TARGET_REQUIRED",

    "PHYSICAL_FILE_LOCAL_EVIDENCE":
        "DETERMINISTIC_LOCAL_UTTERANCE_TARGET_REQUIRED",

    "PHYSICAL_FILE_MULTI_SPEAKER":
        "DETERMINISTIC_ATTRIBUTION_AWARE_LOCAL_TARGET_REQUIRED",

    "ADJACENT_UTTERANCE_PAIR":
        "USE_EXACT_ADJACENT_PAIR",

    "PHYSICAL_FILE_MIXED_AUTHORITY":
        "DETERMINISTIC_AUTHORITY_CONTRAST_TARGET_REQUIRED",

    "SOURCE_UTTERANCE":
        "USE_EXACT_SOURCE_UTTERANCE",

    "SOURCE_SENTENCE_TOKEN_FALLBACK":
        "USE_EXACT_SOURCE_SENTENCE",

    "ADJACENT_OR_LOCAL_TRANSCRIPT_CONTEXT":
        "HUMAN_CONFIRMED_QA_TARGET_REQUIRED",
}


# ============================================================
# 4) Future Task Draft schema
# ============================================================

TASK_DRAFT_SCHEMA_V1_DRAFT = {
    "version":
        "golden_task_draft_v1_draft",

    "status":
        "SCHEMA_DEFINED_NOT_MATERIALIZED",

    "required_fields": [
        "task_draft_id",
        "golden_case_id",
        "source_type",
        "stratum",
        "task_family",
        "question_archetype",
        "evidence_scope",
        "evidence_target_method",
        "required_dimensions",
        "authority_expectation",
        "context_expectation",
        "construction_eligibility",
        "target_evidence_status",
        "question_text_status",
        "expected_answer_status",
        "citation_target_status",
    ],

    "future_target_evidence_fields": [
        "target_document_ids",
        "target_locator_refs",
        "target_text_spans",
        "target_attribution_state",
        "target_authority_state",
        "target_selection_basis",
    ],

    "question_text_rules": {
        "natural_user_question":
            True,

        "internal_document_id_leakage":
            "FORBIDDEN",

        "hash_or_internal_pipeline_metadata_leakage":
            "FORBIDDEN",

        "answer_leakage_in_question":
            "FORBIDDEN",

        "external_knowledge_dependency":
            "FORBIDDEN",

        "unsupported_attribution":
            "FORBIDDEN",

        "unsupported_authority_promotion":
            "FORBIDDEN",

        "unknown_speaker_inference":
            "FORBIDDEN",
    },

    "target_before_question":
        True,

    "question_generation_status":
        "NOT_STARTED",

    "expected_answer_generation_status":
        "NOT_STARTED",

    "artifact_write_status":
        "NONE",
}


# ============================================================
# 5) Stable planning ID
#
# This identifies the future task draft shell.
# It does NOT depend on question text.
# ============================================================

TASK_DRAFT_METHOD_VERSION = (
    "task_draft_method_v1_draft"
)


def build_task_draft_id(
    golden_case_id,
    task_family,
    question_archetype,
):

    payload = {
        "method_version":
            TASK_DRAFT_METHOD_VERSION,

        "golden_case_id":
            golden_case_id,

        "task_family":
            task_family,

        "question_archetype":
            question_archetype,
    }

    digest = hashlib.sha256(
        stable_json_key(
            payload
        ).encode(
            "utf-8"
        )
    ).hexdigest()

    return (
        "task-draft:"
        + digest[
            :32
        ]
    )


# ============================================================
# 6) Build planning shells for all 724 cases
#
# No question text or evidence target is constructed here.
# ============================================================

task_draft_plans = {}

task_draft_plan_rows = []

missing_target_method_cases = []

review_target_leakage = []


for case_record in (
    golden_case_records
):

    case_id = (
        case_record[
            "golden_case_id"
        ]
    )

    source_type = (
        case_record[
            "source_type"
        ]
    )

    stratum = (
        case_record[
            "stratum"
        ]
    )

    task_design = (
        case_task_design[
            case_id
        ]
    )

    question_archetype = (
        QUESTION_ARCHETYPE_BY_STRATUM[
            source_type
        ][
            stratum
        ]
    )

    evidence_scope = (
        task_design[
            "evidence_scope"
        ]
    )

    evidence_target_method = (
        EVIDENCE_TARGET_METHOD_BY_SCOPE.get(
            evidence_scope
        )
    )


    if (
        evidence_target_method
        is None
    ):

        missing_target_method_cases.append(
            {
                "golden_case_id":
                    case_id,

                "source_type":
                    source_type,

                "stratum":
                    stratum,

                "evidence_scope":
                    evidence_scope,
            }
        )


    groundability = (
        case_groundability_precheck[
            case_id
        ]
    )


    if (
        groundability[
            "construction_eligibility"
        ]
        ==
        "READY_FOR_GROUNDED_TASK_DRAFTING"
    ):

        construction_eligibility = (
            "READY_FOR_TARGET_EVIDENCE_SELECTION"
        )

        target_evidence_status = (
            "NOT_SELECTED"
        )

    else:

        construction_eligibility = (
            "BLOCKED_PENDING_HUMAN_CONFIRMATION"
        )

        target_evidence_status = (
            "BLOCKED"
        )


    if (
        question_archetype
        ==
        "REVIEW_CONFIRMATION_ONLY"
    ):

        construction_eligibility = (
            "BLOCKED_PENDING_HUMAN_CONFIRMATION"
        )

        target_evidence_status = (
            "BLOCKED"
        )


    if (
        case_record[
            "final_golden_membership"
        ]
        is False

        and

        target_evidence_status
        != "BLOCKED"
    ):

        review_target_leakage.append(
            case_id
        )


    task_draft_id = (
        build_task_draft_id(
            golden_case_id=(
                case_id
            ),
            task_family=(
                task_design[
                    "task_family"
                ]
            ),
            question_archetype=(
                question_archetype
            ),
        )
    )


    plan = {
        "task_draft_id":
            task_draft_id,

        "golden_case_id":
            case_id,

        "source_type":
            source_type,

        "stratum":
            stratum,

        "task_family":
            task_design[
                "task_family"
            ],

        "question_archetype":
            question_archetype,

        "evidence_scope":
            evidence_scope,

        "evidence_target_method":
            evidence_target_method,

        "required_dimensions":
            list(
                task_design[
                    "required_dimensions"
                ]
            ),

        "authority_expectation":
            task_design[
                "authority_expectation"
            ],

        "context_expectation":
            task_design[
                "context_expectation"
            ],

        "construction_eligibility":
            construction_eligibility,

        "target_evidence_status":
            target_evidence_status,

        "question_text_status":
            "NOT_STARTED",

        "expected_answer_status":
            "NOT_STARTED",

        "citation_target_status":
            "NOT_STARTED",
    }


    task_draft_plans[
        task_draft_id
    ] = plan

    task_draft_plan_rows.append(
        plan
    )


# ============================================================
# 7) Independent planning-ID recomputation
# ============================================================

task_draft_id_failures = []


for plan in (
    task_draft_plan_rows
):

    recomputed = (
        build_task_draft_id(
            golden_case_id=(
                plan[
                    "golden_case_id"
                ]
            ),
            task_family=(
                plan[
                    "task_family"
                ]
            ),
            question_archetype=(
                plan[
                    "question_archetype"
                ]
            ),
        )
    )

    if (
        recomputed
        !=
        plan[
            "task_draft_id"
        ]
    ):

        task_draft_id_failures.append(
            plan[
                "task_draft_id"
            ]
        )


# ============================================================
# 8) Method coverage audits
# ============================================================

archetype_coverage_failures = []

scope_method_failures = []


for source_type in [
    "website",
    "book",
    "transcript",
]:

    expected = set(
        expected_strata[
            source_type
        ]
    )

    actual = set(
        QUESTION_ARCHETYPE_BY_STRATUM[
            source_type
        ].keys()
    )

    if (
        expected
        != actual
    ):

        archetype_coverage_failures.append(
            {
                "source_type":
                    source_type,

                "missing":
                    sorted(
                        expected - actual
                    ),

                "extra":
                    sorted(
                        actual - expected
                    ),
            }
        )


for row in (
    task_design_rows
):

    if (
        row[
            "evidence_scope"
        ]
        not in
        EVIDENCE_TARGET_METHOD_BY_SCOPE
    ):

        scope_method_failures.append(
            {
                "source_type":
                    row[
                        "source_type"
                    ],

                "stratum":
                    row[
                        "stratum"
                    ],

                "evidence_scope":
                    row[
                        "evidence_scope"
                    ],
            }
        )


# ============================================================
# 9) Summary counts
# ============================================================

ready_target_selection_count = sum(
    1
    for plan
    in task_draft_plan_rows
    if (
        plan[
            "construction_eligibility"
        ]
        ==
        "READY_FOR_TARGET_EVIDENCE_SELECTION"
    )
)


blocked_task_plan_count = sum(
    1
    for plan
    in task_draft_plan_rows
    if (
        plan[
            "construction_eligibility"
        ]
        ==
        "BLOCKED_PENDING_HUMAN_CONFIRMATION"
    )
)


archetype_usage_counts = Counter(
    plan[
        "question_archetype"
    ]
    for plan
    in task_draft_plan_rows
)


target_method_usage_counts = Counter(
    plan[
        "evidence_target_method"
    ]
    for plan
    in task_draft_plan_rows
)


# ============================================================
# 10) Construction Method Contract
# ============================================================

QUESTION_CONSTRUCTION_METHOD_V1_DRAFT = {
    "version":
        TASK_DRAFT_METHOD_VERSION,

    "status":
        "VALIDATED_IN_MEMORY_NOT_EXECUTED",

    "case_count":
        len(
            task_draft_plan_rows
        ),

    "ready_for_target_evidence_selection":
        ready_target_selection_count,

    "blocked_pending_human_confirmation":
        blocked_task_plan_count,

    "sequence": [
        "CASE_VALIDATED",
        "TASK_DESIGN_VALIDATED",
        "GROUNDABILITY_PRECHECK_PASS",
        "TARGET_EVIDENCE_SELECTION",
        "TARGET_EVIDENCE_VALIDATION",
        "QUESTION_DRAFT",
        "QUESTION_GROUNDABILITY_VALIDATION",
        "EXPECTED_ANSWER_CONSTRUCTION",
        "CITATION_TARGET_VALIDATION",
    ],

    "critical_rule":
        "TARGET_EVIDENCE_MUST_BE_SELECTED_AND_VALIDATED_BEFORE_QUESTION_TEXT",

    "question_generation_status":
        "NOT_STARTED",

    "target_evidence_selection_status":
        "NOT_STARTED",

    "expected_answer_generation_status":
        "NOT_STARTED",

    "artifact_write_status":
        "NONE",
}


# ============================================================
# 11) Validation gates
# ============================================================

exact_27_archetype_mapping_gate = (
    len(
        archetype_coverage_failures
    )
    == 0
)


all_evidence_scopes_have_target_method_gate = (
    len(
        scope_method_failures
    )
    == 0

    and

    len(
        missing_target_method_cases
    )
    == 0
)


all_724_task_plans_gate = (
    len(
        task_draft_plan_rows
    )
    ==
    len(
        golden_case_records
    )
    ==
    724
)


task_draft_ids_unique_gate = (
    len(
        task_draft_plans
    )
    ==
    len(
        task_draft_plan_rows
    )
)


task_draft_ids_recomputable_gate = (
    len(
        task_draft_id_failures
    )
    == 0
)


exact_572_ready_for_target_selection_gate = (
    ready_target_selection_count
    == 572
)


exact_152_blocked_gate = (
    blocked_task_plan_count
    == 152
)


no_review_target_leakage_gate = (
    len(
        review_target_leakage
    )
    == 0
)


review_archetype_blocked_gate = all(
    plan[
        "construction_eligibility"
    ]
    ==
    "BLOCKED_PENDING_HUMAN_CONFIRMATION"

    for plan
    in task_draft_plan_rows

    if (
        plan[
            "question_archetype"
        ]
        ==
        "REVIEW_CONFIRMATION_ONLY"
    )
)


target_before_question_gate = (
    TASK_DRAFT_SCHEMA_V1_DRAFT[
        "target_before_question"
    ]
    is True

    and

    QUESTION_CONSTRUCTION_METHOD_V1_DRAFT[
        "critical_rule"
    ]
    ==
    "TARGET_EVIDENCE_MUST_BE_SELECTED_AND_VALIDATED_BEFORE_QUESTION_TEXT"
)


no_internal_metadata_leakage_gate = (
    TASK_DRAFT_SCHEMA_V1_DRAFT[
        "question_text_rules"
    ][
        "internal_document_id_leakage"
    ]
    ==
    "FORBIDDEN"

    and

    TASK_DRAFT_SCHEMA_V1_DRAFT[
        "question_text_rules"
    ][
        "hash_or_internal_pipeline_metadata_leakage"
    ]
    ==
    "FORBIDDEN"
)


unknown_speaker_guardrail_gate = (
    TASK_DRAFT_SCHEMA_V1_DRAFT[
        "question_text_rules"
    ][
        "unknown_speaker_inference"
    ]
    ==
    "FORBIDDEN"
)


question_generation_not_started_gate = (
    QUESTION_CONSTRUCTION_METHOD_V1_DRAFT[
        "question_generation_status"
    ]
    ==
    "NOT_STARTED"

    and

    all(
        plan[
            "question_text_status"
        ]
        ==
        "NOT_STARTED"

        for plan
        in task_draft_plan_rows
    )
)


target_selection_not_started_gate = (
    QUESTION_CONSTRUCTION_METHOD_V1_DRAFT[
        "target_evidence_selection_status"
    ]
    ==
    "NOT_STARTED"

    and

    all(
        plan[
            "target_evidence_status"
        ]
        in {
            "NOT_SELECTED",
            "BLOCKED",
        }

        for plan
        in task_draft_plan_rows
    )
)


expected_answer_not_started_gate = (
    QUESTION_CONSTRUCTION_METHOD_V1_DRAFT[
        "expected_answer_generation_status"
    ]
    ==
    "NOT_STARTED"

    and

    all(
        plan[
            "expected_answer_status"
        ]
        ==
        "NOT_STARTED"

        for plan
        in task_draft_plan_rows
    )
)


no_artifact_write_gate = (
    TASK_DRAFT_SCHEMA_V1_DRAFT[
        "artifact_write_status"
    ]
    == "NONE"

    and

    QUESTION_CONSTRUCTION_METHOD_V1_DRAFT[
        "artifact_write_status"
    ]
    == "NONE"
)


cell_26_gates = {
    "exact_27_strata_have_question_archetype":
        exact_27_archetype_mapping_gate,

    "all_evidence_scopes_have_target_selection_method":
        all_evidence_scopes_have_target_method_gate,

    "all_724_cases_have_task_draft_plan":
        all_724_task_plans_gate,

    "task_draft_ids_unique":
        task_draft_ids_unique_gate,

    "task_draft_ids_independently_recomputable":
        task_draft_ids_recomputable_gate,

    "exact_572_ready_for_target_evidence_selection":
        exact_572_ready_for_target_selection_gate,

    "exact_152_review_cases_remain_blocked":
        exact_152_blocked_gate,

    "no_review_case_leaks_into_target_selection":
        no_review_target_leakage_gate,

    "review_only_archetypes_remain_blocked":
        review_archetype_blocked_gate,

    "target_evidence_required_before_question":
        target_before_question_gate,

    "internal_ids_and_hashes_forbidden_in_questions":
        no_internal_metadata_leakage_gate,

    "unknown_speaker_inference_forbidden":
        unknown_speaker_guardrail_gate,

    "target_evidence_selection_not_started":
        target_selection_not_started_gate,

    "question_generation_not_started":
        question_generation_not_started_gate,

    "expected_answer_generation_not_started":
        expected_answer_not_started_gate,

    "no_artifact_write":
        no_artifact_write_gate,
}


cell_26_pass = all(
    cell_26_gates.values()
)


# ============================================================
# 12) Output
# ============================================================

print(
    "=== TASK DRAFT SCHEMA & QUESTION CONSTRUCTION METHOD ==="
)

print(
    "Task draft plans:",
    len(
        task_draft_plan_rows
    )
)

print(
    "Ready for target evidence selection:",
    ready_target_selection_count
)

print(
    "Blocked pending human confirmation:",
    blocked_task_plan_count
)


print(
    "\n=== QUESTION ARCHETYPE USAGE ==="
)

for archetype, count in sorted(
    archetype_usage_counts.items()
):

    print(
        "-",
        archetype,
        ":",
        count,
    )


print(
    "\n=== EVIDENCE TARGET METHOD USAGE ==="
)

for method, count in sorted(
    target_method_usage_counts.items(),
    key=lambda item:
        str(
            item[
                0
            ]
        ),
):

    print(
        "-",
        method,
        ":",
        count,
    )


print(
    "\n=== CONSTRUCTION SEQUENCE ==="
)

for index, step in enumerate(
    QUESTION_CONSTRUCTION_METHOD_V1_DRAFT[
        "sequence"
    ],
    start=1,
):

    print(
        index,
        step,
    )


print(
    "\n=== REPRESENTATIVE TASK DRAFT PLANS ==="
)

for plan in (
    task_draft_plan_rows[
        :8
    ]
):

    print(
        json.dumps(
            {
                "task_draft_id":
                    plan[
                        "task_draft_id"
                    ],

                "golden_case_id":
                    plan[
                        "golden_case_id"
                    ],

                "source_type":
                    plan[
                        "source_type"
                    ],

                "stratum":
                    plan[
                        "stratum"
                    ],

                "task_family":
                    plan[
                        "task_family"
                    ],

                "question_archetype":
                    plan[
                        "question_archetype"
                    ],

                "evidence_scope":
                    plan[
                        "evidence_scope"
                    ],

                "evidence_target_method":
                    plan[
                        "evidence_target_method"
                    ],

                "construction_eligibility":
                    plan[
                        "construction_eligibility"
                    ],

                "target_evidence_status":
                    plan[
                        "target_evidence_status"
                    ],

                "question_text_status":
                    plan[
                        "question_text_status"
                    ],
            },
            ensure_ascii=False,
            sort_keys=True,
        )
    )


print(
    "\n=== CODE CELL 26 GATES ==="
)

for name, passed in (
    cell_26_gates.items()
):

    print(
        name,
        ":",
        "PASS"
        if passed
        else "FAIL"
    )


print(
    "\nCODE CELL 26 GATE:",
    "PASS"
    if cell_26_pass
    else "FAIL"
)


assert cell_26_pass, (
    "Task Draft Schema / Question Construction Method validation failed. "
    "Do not select target evidence or generate question text until "
    "all case plans, archetypes, target methods, and guardrails pass."
)
```

<h3>Target Evidence Selection Readiness Audit</h3>

<p>
۵۷۲ Case اکنون برای Target Evidence Selection آماده‌اند؛
اما قبل از انتخاب واقعی Target باید ثابت شود که هر
<code>evidence_target_method</code>
به اطلاعات واقعی و کافی دسترسی دارد.
</p>

<p>
در این مرحله هیچ Target انتخاب نمی‌شود.
فقط Candidate Snapshot، Evidence Documentها و Metadata موجود برای هر Method بررسی می‌شوند.
</p>

<p>
هدف این است که مشخص شود برای Methodهایی مانند:

<p>
چه Fieldهایی واقعاً قابل استفاده هستند و کدام Method به Reconstruction بیشتری نیاز دارد.
</p>

<p>
هیچ First-record shortcut، Random choice یا Heuristic حدسی مجاز نیست.
اگر Methodی Evidence کافی نداشته باشد، باید همان Method Repair شود و نه اینکه Target دلخواه ساخته شود.
</p>

<p>
Question Generation و Expected Answer Generation همچنان
<code>NOT_STARTED</code>

```python
# Code Cell 27

# ============================================================
# Target Evidence Selection Readiness Audit
#
# INSPECTION ONLY.
#
# NO target evidence selection.
# NO question generation.
# NO expected-answer generation.
# NO artifact writes.
# ============================================================


# ============================================================
# 1) Full accepted-corpus lightweight record index
# ============================================================

corpus_target_index = {}


with corpus_path.open(
    "r",
    encoding="utf-8",
) as handle:

    for line in handle:

        if not line.strip():
            continue

        record = json.loads(
            line
        )

        document_id = (
            record.get(
                "document_id"
            )
        )

        metadata = (
            record.get(
                "source_metadata"
            )
            or {}
        )

        locator = (
            record.get(
                "locator"
            )
            or {}
        )

        corpus_target_index[
            document_id
        ] = {
            "document_id":
                document_id,

            "source_id":
                record.get(
                    "source_id"
                ),

            "source_type":
                record.get(
                    "source_type"
                ),

            "text":
                str(
                    record.get(
                        "text"
                    )
                    or ""
                ),

            "title":
                record.get(
                    "title"
                ),

            "source_metadata":
                metadata,

            "locator":
                locator,

            "attribution":
                record.get(
                    "attribution"
                ),

            "rights":
                record.get(
                    "rights"
                ),

            "lineage":
                record.get(
                    "lineage"
                ),
        }


# ============================================================
# 2) Method-level required evidence capabilities
#
# These are capability requirements, not selection rules yet.
# ============================================================

TARGET_METHOD_CAPABILITY_REQUIREMENTS = {

    "DOCUMENT_LEVEL_TARGET": {
        "requires_single_document":
            True,

        "requires_local_span_selection":
            False,

        "requires_multi_record_structure":
            False,

        "requires_attribution_signal":
            False,

        "requires_exact_candidate_span":
            False,
    },


    "DETERMINISTIC_LOCAL_EVIDENCE_TARGET_REQUIRED": {
        "requires_single_document":
            True,

        "requires_local_span_selection":
            True,

        "requires_multi_record_structure":
            False,

        "requires_attribution_signal":
            False,

        "requires_exact_candidate_span":
            False,
    },


    "USE_EXACT_PROTECTED_REGION": {
        "requires_single_document":
            False,

        "requires_local_span_selection":
            False,

        "requires_multi_record_structure":
            True,

        "requires_attribution_signal":
            False,

        "requires_exact_candidate_span":
            True,
    },


    "USE_EXACT_PROTECTED_TABLE_REGION": {
        "requires_single_document":
            False,

        "requires_local_span_selection":
            False,

        "requires_multi_record_structure":
            True,

        "requires_attribution_signal":
            False,

        "requires_exact_candidate_span":
            True,
    },


    "DETERMINISTIC_LOCAL_OR_MULTI_RECORD_TARGET_REQUIRED": {
        "requires_single_document":
            False,

        "requires_local_span_selection":
            True,

        "requires_multi_record_structure":
            True,

        "requires_attribution_signal":
            False,

        "requires_exact_candidate_span":
            False,
    },


    "USE_EXACT_TABLE_RUN": {
        "requires_single_document":
            False,

        "requires_local_span_selection":
            False,

        "requires_multi_record_structure":
            True,

        "requires_attribution_signal":
            False,

        "requires_exact_candidate_span":
            True,
    },


    "USE_EXPLICIT_BOOK_REGION": {
        "requires_single_document":
            False,

        "requires_local_span_selection":
            False,

        "requires_multi_record_structure":
            False,

        "requires_attribution_signal":
            False,

        "requires_exact_candidate_span":
            True,
    },


    "USE_PARENT_PLUS_SUPPLEMENTAL_RELATION": {
        "requires_single_document":
            False,

        "requires_local_span_selection":
            False,

        "requires_multi_record_structure":
            True,

        "requires_attribution_signal":
            False,

        "requires_exact_candidate_span":
            True,
    },


    "USE_RECORD_PLUS_HEADING_PATH": {
        "requires_single_document":
            True,

        "requires_local_span_selection":
            False,

        "requires_multi_record_structure":
            False,

        "requires_attribution_signal":
            False,

        "requires_exact_candidate_span":
            True,
    },


    "DETERMINISTIC_LOCAL_UTTERANCE_TARGET_REQUIRED": {
        "requires_single_document":
            False,

        "requires_local_span_selection":
            True,

        "requires_multi_record_structure":
            True,

        "requires_attribution_signal":
            True,

        "requires_exact_candidate_span":
            False,
    },


    "DETERMINISTIC_ATTRIBUTION_AWARE_LOCAL_TARGET_REQUIRED": {
        "requires_single_document":
            False,

        "requires_local_span_selection":
            True,

        "requires_multi_record_structure":
            True,

        "requires_attribution_signal":
            True,

        "requires_exact_candidate_span":
            False,
    },


    "USE_EXACT_ADJACENT_PAIR": {
        "requires_single_document":
            False,

        "requires_local_span_selection":
            False,

        "requires_multi_record_structure":
            True,

        "requires_attribution_signal":
            True,

        "requires_exact_candidate_span":
            True,
    },


    "DETERMINISTIC_AUTHORITY_CONTRAST_TARGET_REQUIRED": {
        "requires_single_document":
            False,

        "requires_local_span_selection":
            True,

        "requires_multi_record_structure":
            True,

        "requires_attribution_signal":
            True,

        "requires_exact_candidate_span":
            False,
    },


    "USE_EXACT_SOURCE_UTTERANCE": {
        "requires_single_document":
            True,

        "requires_local_span_selection":
            False,

        "requires_multi_record_structure":
            False,

        "requires_attribution_signal":
            False,

        "requires_exact_candidate_span":
            True,
    },


    "USE_EXACT_SOURCE_SENTENCE": {
        "requires_single_document":
            True,

        "requires_local_span_selection":
            False,

        "requires_multi_record_structure":
            False,

        "requires_attribution_signal":
            False,

        "requires_exact_candidate_span":
            True,
    },


    # Review-only methods remain blocked.
    "HUMAN_CONFIRMED_LOCAL_TARGET_REQUIRED": {
        "review_only":
            True,
    },

    "HUMAN_CONFIRMED_SOURCE_AND_CONTRAST_TARGET_REQUIRED": {
        "review_only":
            True,
    },

    "HUMAN_CONFIRMED_MULTI_PARAGRAPH_TARGET_REQUIRED": {
        "review_only":
            True,
    },

    "HUMAN_CONFIRMED_QA_TARGET_REQUIRED": {
        "review_only":
            True,
    },
}


# ============================================================
# 3) Transcript evidence-order reconstruction
# ============================================================

transcript_records_by_physical_file = defaultdict(
    list
)


for document_id, record in (
    corpus_target_index.items()
):

    if (
        record[
            "source_type"
        ]
        != "transcript"
    ):
        continue

    locator = (
        record[
            "locator"
        ]
        or {}
    )

    metadata = (
        record[
            "source_metadata"
        ]
        or {}
    )

    physical_file_id = (
        locator.get(
            "physical_file_id"
        )
    )

    if (
        physical_file_id
        is None
    ):
        continue

    transcript_records_by_physical_file[
        physical_file_id
    ].append(
        {
            "document_id":
                document_id,

            "utterance_index":
                metadata.get(
                    "utterance_index"
                ),

            "text":
                record[
                    "text"
                ],

            "source_id":
                record[
                    "source_id"
                ],

            "source_metadata":
                metadata,

            "locator":
                locator,

            "attribution":
                record[
                    "attribution"
                ],
        }
    )


for physical_file_id in list(
    transcript_records_by_physical_file
):

    transcript_records_by_physical_file[
        physical_file_id
    ] = sorted(
        transcript_records_by_physical_file[
            physical_file_id
        ],
        key=lambda item: (
            item[
                "utterance_index"
            ]
            if (
                item[
                    "utterance_index"
                ]
                is not None
            )
            else 10**12
        ),
    )


# ============================================================
# 4) Inspect actual ready cases by target method
# ============================================================

target_method_readiness = defaultdict(
    lambda: {
        "case_count":
            0,

        "evidence_document_counts":
            Counter(),

        "candidate_key_counts":
            Counter(),

        "missing_candidate_fields":
            Counter(),

        "missing_record_support_count":
            0,

        "attribution_signal_missing_count":
            0,

        "local_structure_missing_count":
            0,

        "examples":
            [],
    }
)


READY_METHOD_REQUIRED_FIELDS = {

    "DOCUMENT_LEVEL_TARGET": [
        "document_id",
    ],

    "DETERMINISTIC_LOCAL_EVIDENCE_TARGET_REQUIRED": [
        "document_id",
    ],

    "USE_EXACT_PROTECTED_REGION": [
        "member_document_ids",
        "start_paragraph",
        "end_paragraph",
        "heading_path",
    ],

    "USE_EXACT_PROTECTED_TABLE_REGION": [
        "member_document_ids",
        "start_paragraph",
        "end_paragraph",
        "heading_path",
    ],

    "DETERMINISTIC_LOCAL_OR_MULTI_RECORD_TARGET_REQUIRED": [
        "document_ids",
        "start_paragraph",
        "end_paragraph",
        "heading_path",
    ],

    "USE_EXACT_TABLE_RUN": [
        "document_ids",
        "start_paragraph",
        "end_paragraph",
    ],

    "USE_EXPLICIT_BOOK_REGION": [
        "document_id",
    ],

    "USE_PARENT_PLUS_SUPPLEMENTAL_RELATION": [
        "document_id",
        "parent_paragraph_id",
    ],

    "USE_RECORD_PLUS_HEADING_PATH": [
        "document_id",
    ],

    "DETERMINISTIC_LOCAL_UTTERANCE_TARGET_REQUIRED": [
        "physical_file_id",
    ],

    "DETERMINISTIC_ATTRIBUTION_AWARE_LOCAL_TARGET_REQUIRED": [
        "physical_file_id",
    ],

    "USE_EXACT_ADJACENT_PAIR": [
        "physical_file_id",
        "left_document_id",
        "right_document_id",
    ],

    "DETERMINISTIC_AUTHORITY_CONTRAST_TARGET_REQUIRED": [
        "physical_file_id",
    ],

    "USE_EXACT_SOURCE_UTTERANCE": [
        "document_id",
    ],

    "USE_EXACT_SOURCE_SENTENCE": [
        "document_id",
        "char_start",
        "char_end",
        "sentence_index",
    ],
}


ready_case_plans = [
    plan
    for plan
    in task_draft_plan_rows
    if (
        plan[
            "construction_eligibility"
        ]
        ==
        "READY_FOR_TARGET_EVIDENCE_SELECTION"
    )
]


for plan in (
    ready_case_plans
):

    case_id = (
        plan[
            "golden_case_id"
        ]
    )

    method = (
        plan[
            "evidence_target_method"
        ]
    )

    case_record = (
        golden_case_records_by_id[
            case_id
        ]
    )

    candidate = (
        case_record[
            "candidate_snapshot"
        ]
    )

    state = (
        target_method_readiness[
            method
        ]
    )

    state[
        "case_count"
    ] += 1

    state[
        "evidence_document_counts"
    ][
        case_record[
            "evidence_document_count"
        ]
    ] += 1


    for key in (
        candidate.keys()
    ):

        state[
            "candidate_key_counts"
        ][
            key
        ] += 1


    required_fields = (
        READY_METHOD_REQUIRED_FIELDS[
            method
        ]
    )


    missing_fields = [
        field
        for field
        in required_fields
        if (
            candidate.get(
                field
            )
            is None
        )
    ]


    for field in (
        missing_fields
    ):

        state[
            "missing_candidate_fields"
        ][
            field
        ] += 1


    # --------------------------------------------------------
    # Every currently resolved evidence document must exist.
    # --------------------------------------------------------

    if not all(
        document_id
        in corpus_target_index

        for document_id
        in case_record[
            "evidence_document_ids"
        ]
    ):

        state[
            "missing_record_support_count"
        ] += 1


    # --------------------------------------------------------
    # Transcript-local methods require physical-file structure.
    # --------------------------------------------------------

    if method in {
        "DETERMINISTIC_LOCAL_UTTERANCE_TARGET_REQUIRED",
        "DETERMINISTIC_ATTRIBUTION_AWARE_LOCAL_TARGET_REQUIRED",
        "DETERMINISTIC_AUTHORITY_CONTRAST_TARGET_REQUIRED",
    }:

        physical_file_id = (
            candidate.get(
                "physical_file_id"
            )
        )

        local_records = (
            transcript_records_by_physical_file.get(
                physical_file_id,
                [],
            )
        )

        if not local_records:

            state[
                "local_structure_missing_count"
            ] += 1


        if method in {
            "DETERMINISTIC_ATTRIBUTION_AWARE_LOCAL_TARGET_REQUIRED",
            "DETERMINISTIC_AUTHORITY_CONTRAST_TARGET_REQUIRED",
        }:

            attribution_signal_count = 0

            for local_record in (
                local_records
            ):

                metadata = (
                    local_record[
                        "source_metadata"
                    ]
                    or {}
                )

                speaker = (
                    metadata.get(
                        "speaker"
                    )
                    or {}
                )

                if (
                    speaker.get(
                        "base"
                    )
                    or
                    speaker.get(
                        "effective"
                    )
                    or
                    local_record.get(
                        "attribution"
                    )
                ):

                    attribution_signal_count += 1


            if (
                attribution_signal_count
                == 0
            ):

                state[
                    "attribution_signal_missing_count"
                ] += 1


    # --------------------------------------------------------
    # Representative examples
    # --------------------------------------------------------

    if (
        len(
            state[
                "examples"
            ]
        )
        < 3
    ):

        state[
            "examples"
        ].append(
            {
                "golden_case_id":
                    case_id,

                "source_type":
                    plan[
                        "source_type"
                    ],

                "stratum":
                    plan[
                        "stratum"
                    ],

                "evidence_document_count":
                    case_record[
                        "evidence_document_count"
                    ],

                "candidate_keys":
                    sorted(
                        candidate.keys()
                    ),

                "missing_required_fields":
                    missing_fields,
            }
        )


# ============================================================
# 5) Review-only method audit
# ============================================================

review_method_case_counts = Counter()

review_method_leakage = []


for plan in (
    task_draft_plan_rows
):

    method = (
        plan[
            "evidence_target_method"
        ]
    )

    requirements = (
        TARGET_METHOD_CAPABILITY_REQUIREMENTS[
            method
        ]
    )

    if not requirements.get(
        "review_only"
    ):
        continue

    review_method_case_counts[
        method
    ] += 1

    if (
        plan[
            "construction_eligibility"
        ]
        !=
        "BLOCKED_PENDING_HUMAN_CONFIRMATION"

        or

        plan[
            "target_evidence_status"
        ]
        != "BLOCKED"
    ):

        review_method_leakage.append(
            plan[
                "task_draft_id"
            ]
        )


# ============================================================
# 6) Determine readiness status per executable method
# ============================================================

target_method_status_rows = []


for method, requirements in (
    TARGET_METHOD_CAPABILITY_REQUIREMENTS.items()
):

    if requirements.get(
        "review_only"
    ):

        target_method_status_rows.append(
            {
                "method":
                    method,

                "case_count":
                    review_method_case_counts[
                        method
                    ],

                "status":
                    "BLOCKED_BY_DESIGN_PENDING_HUMAN_REVIEW",
            }
        )

        continue


    state = (
        target_method_readiness[
            method
        ]
    )


    missing_required_count = sum(
        state[
            "missing_candidate_fields"
        ].values()
    )


    ready = (
        state[
            "case_count"
        ]
        > 0

        and

        missing_required_count
        == 0

        and

        state[
            "missing_record_support_count"
        ]
        == 0

        and

        state[
            "local_structure_missing_count"
        ]
        == 0

        and

        state[
            "attribution_signal_missing_count"
        ]
        == 0
    )


    target_method_status_rows.append(
        {
            "method":
                method,

            "case_count":
                state[
                    "case_count"
                ],

            "status":
                (
                    "READY_FOR_DETERMINISTIC_SELECTION"
                    if ready
                    else
                    "NOT_READY_REQUIRES_METHOD_REPAIR"
                ),

            "missing_required_field_count":
                missing_required_count,

            "missing_record_support_count":
                state[
                    "missing_record_support_count"
                ],

            "local_structure_missing_count":
                state[
                    "local_structure_missing_count"
                ],

            "attribution_signal_missing_count":
                state[
                    "attribution_signal_missing_count"
                ],
        }
    )


# ============================================================
# 7) Readiness contract
# ============================================================

TARGET_EVIDENCE_SELECTION_READINESS_V1_DRAFT = {
    "version":
        "target_evidence_selection_readiness_v1_draft",

    "status":
        "INSPECTED_NOT_EXECUTED",

    "ready_case_count":
        len(
            ready_case_plans
        ),

    "review_blocked_case_count":
        blocked_task_plan_count,

    "method_statuses":
        target_method_status_rows,

    "selection_status":
        "NOT_STARTED",

    "question_generation_status":
        "NOT_STARTED",

    "expected_answer_generation_status":
        "NOT_STARTED",

    "artifact_write_status":
        "NONE",
}


# ============================================================
# 8) Validation gates
# ============================================================

exact_572_ready_cases_inspected_gate = (
    len(
        ready_case_plans
    )
    == 572
)


all_executable_methods_ready_gate = all(
    row[
        "status"
    ]
    ==
    "READY_FOR_DETERMINISTIC_SELECTION"

    for row
    in target_method_status_rows

    if (
        row[
            "method"
        ]
        not in {
            "HUMAN_CONFIRMED_LOCAL_TARGET_REQUIRED",
            "HUMAN_CONFIRMED_SOURCE_AND_CONTRAST_TARGET_REQUIRED",
            "HUMAN_CONFIRMED_MULTI_PARAGRAPH_TARGET_REQUIRED",
            "HUMAN_CONFIRMED_QA_TARGET_REQUIRED",
        }
    )
)


all_review_methods_blocked_gate = all(
    row[
        "status"
    ]
    ==
    "BLOCKED_BY_DESIGN_PENDING_HUMAN_REVIEW"

    for row
    in target_method_status_rows

    if (
        row[
            "method"
        ]
        in {
            "HUMAN_CONFIRMED_LOCAL_TARGET_REQUIRED",
            "HUMAN_CONFIRMED_SOURCE_AND_CONTRAST_TARGET_REQUIRED",
            "HUMAN_CONFIRMED_MULTI_PARAGRAPH_TARGET_REQUIRED",
            "HUMAN_CONFIRMED_QA_TARGET_REQUIRED",
        }
    )
)


exact_152_review_cases_gate = (
    sum(
        review_method_case_counts.values()
    )
    == 152
)


no_review_method_leakage_gate = (
    len(
        review_method_leakage
    )
    == 0
)


all_ready_case_evidence_documents_exist_gate = all(
    document_id
    in corpus_target_index

    for plan
    in ready_case_plans

    for document_id
    in golden_case_records_by_id[
        plan[
            "golden_case_id"
        ]
    ][
        "evidence_document_ids"
    ]
)


target_selection_not_started_gate = (
    TARGET_EVIDENCE_SELECTION_READINESS_V1_DRAFT[
        "selection_status"
    ]
    == "NOT_STARTED"

    and

    all(
        plan[
            "target_evidence_status"
        ]
        in {
            "NOT_SELECTED",
            "BLOCKED",
        }

        for plan
        in task_draft_plan_rows
    )
)


question_generation_not_started_gate = (
    TARGET_EVIDENCE_SELECTION_READINESS_V1_DRAFT[
        "question_generation_status"
    ]
    == "NOT_STARTED"

    and

    all(
        plan[
            "question_text_status"
        ]
        == "NOT_STARTED"

        for plan
        in task_draft_plan_rows
    )
)


expected_answer_not_started_gate = (
    TARGET_EVIDENCE_SELECTION_READINESS_V1_DRAFT[
        "expected_answer_generation_status"
    ]
    == "NOT_STARTED"

    and

    all(
        plan[
            "expected_answer_status"
        ]
        == "NOT_STARTED"

        for plan
        in task_draft_plan_rows
    )
)


no_artifact_write_gate = (
    TARGET_EVIDENCE_SELECTION_READINESS_V1_DRAFT[
        "artifact_write_status"
    ]
    == "NONE"
)


cell_27_gates = {
    "exact_572_ready_cases_inspected":
        exact_572_ready_cases_inspected_gate,

    "all_executable_target_methods_ready":
        all_executable_methods_ready_gate,

    "all_human_confirmed_methods_remain_blocked":
        all_review_methods_blocked_gate,

    "exact_152_review_cases_accounted_for":
        exact_152_review_cases_gate,

    "no_review_method_leakage":
        no_review_method_leakage_gate,

    "all_ready_case_evidence_documents_exist":
        all_ready_case_evidence_documents_exist_gate,

    "target_evidence_selection_not_started":
        target_selection_not_started_gate,

    "question_generation_not_started":
        question_generation_not_started_gate,

    "expected_answer_generation_not_started":
        expected_answer_not_started_gate,

    "no_artifact_write":
        no_artifact_write_gate,
}


cell_27_pass = all(
    cell_27_gates.values()
)


# ============================================================
# 9) Output
# ============================================================

print(
    "=== TARGET EVIDENCE SELECTION READINESS ==="
)

print(
    "Ready cases inspected:",
    len(
        ready_case_plans
    )
)

print(
    "Review-blocked cases:",
    sum(
        review_method_case_counts.values()
    )
)


print(
    "\n=== METHOD READINESS ==="
)

for row in (
    target_method_status_rows
):

    print(
        "-",
        row[
            "method"
        ],
        "| cases:",
        row[
            "case_count"
        ],
        "| status:",
        row[
            "status"
        ],
    )

    for field in [
        "missing_required_field_count",
        "missing_record_support_count",
        "local_structure_missing_count",
        "attribution_signal_missing_count",
    ]:

        if field in row:

            print(
                "   ",
                field,
                ":",
                row[
                    field
                ],
            )


print(
    "\n=== REPRESENTATIVE METHOD INPUTS ==="
)

for method, state in (
    target_method_readiness.items()
):

    if (
        state[
            "case_count"
        ]
        == 0
    ):
        continue

    print(
        "\n[",
        method,
        "]",
        sep="",
    )

    print(
        "cases:",
        state[
            "case_count"
        ]
    )

    print(
        "evidence_document_count distribution:",
        dict(
            sorted(
                state[
                    "evidence_document_counts"
                ].items()
            )
        )
    )

    print(
        "top candidate keys:",
        state[
            "candidate_key_counts"
        ].most_common(
            15
        )
    )

    print(
        "missing required fields:",
        dict(
            state[
                "missing_candidate_fields"
            ]
        )
    )

    for example in (
        state[
            "examples"
        ]
    ):

        print(
            json.dumps(
                example,
                ensure_ascii=False,
                sort_keys=True,
            )
        )


print(
    "\n=== CODE CELL 27 GATES ==="
)

for name, passed in (
    cell_27_gates.items()
):

    print(
        name,
        ":",
        "PASS"
        if passed
        else "FAIL"
    )


print(
    "\nCODE CELL 27 GATE:",
    "PASS"
    if cell_27_pass
    else "FAIL"
)


assert cell_27_pass, (
    "Target Evidence Selection readiness audit failed. "
    "Do not select target evidence until every executable method "
    "has sufficient real candidate/evidence structure."
)
```

<div style="direction: rtl; white-space: normal; line-height: 1.7;">

<h3>Deterministic Target Evidence Selection and Validation</h3>

<p>
Readiness Audit نشان داد تمام ۱۵ Method اجرایی برای ۵۷۲ Case آماده‌اند.
در این مرحله برای اولین بار <code>Target Evidence</code> واقعی هر Case ساخته می‌شود.
</p>

<p>
Target Evidence با Case Evidence یکسان نیست.
Case Evidence می‌تواند یک فایل کامل، Section بزرگ یا مجموعه‌ای از رکوردها باشد؛
اما Target مشخص می‌کند سؤال آینده دقیقاً باید درباره‌ی کدام Evidence ساخته شود.
</p>

<h4>اصول Selection</h4>

<ul>
<li>Methodهای <strong>Exact</strong> همان Region، Table Run، Utterance، Sentence یا Pair فریز‌شده را بدون تغییر استفاده می‌کنند.</li>
<li>Long Websiteها با یک Window محلی و deterministic از Sentenceها هدف‌گذاری می‌شوند.</li>
<li>Long Book Sectionها با کوچک‌ترین Multi-record Window یعنی دو رکورد متوالی و بیشترین Evidence Payload هدف‌گذاری می‌شوند.</li>
<li>Physical Fileهای ساده با یک Utterance واقعی و deterministic هدف‌گذاری می‌شوند.</li>
<li>Multi-speaker Fileها با نزدیک‌ترین Pair دارای Attribution State متفاوت هدف‌گذاری می‌شوند.</li>
<li>Mixed-authority Fileها با نزدیک‌ترین Pair دارای Primary State متفاوت هدف‌گذاری می‌شوند.</li>
<li>هیچ Randomness، First-record shortcut یا Speaker/Authority Inference استفاده نمی‌شود.</li>
<li>۱۵۲ Review Case همچنان Target دریافت نمی‌کنند.</li>
</ul>

<p>
هر Target دارای Stable ID، Document/Span Identity، Locator Reference،
Attribution Evidence و Selection Basis خواهد بود و باید با اجرای مستقل دوباره
دقیقاً بازتولید شود.
</p>

<p>
هنوز Question Text، Expected Answer و Retrieval Judgment تولید نمی‌شود.
</p>

</div>

```python
# Code Cell 28

# ============================================================
# Deterministic Target Evidence Selection
#
# Executes target selection for the 572 ready cases.
#
# NO question generation.
# NO expected-answer generation.
# NO retrieval execution.
# NO artifact writes.
# ============================================================


# ============================================================
# 1) Stable helpers
# ============================================================

TARGET_EVIDENCE_SCHEMA_VERSION = (
    "target_evidence_v1_draft"
)


def target_part_from_document(
    document_id,
    char_start=None,
    char_end=None,
    role=None,
):

    record = (
        corpus_target_index[
            document_id
        ]
    )

    text = (
        record[
            "text"
        ]
    )

    if char_start is None:
        char_start = 0

    if char_end is None:
        char_end = len(
            text
        )

    assert (
        0
        <= char_start
        < char_end
        <= len(
            text
        )
    )

    target_text = text[
        char_start:
        char_end
    ]

    target_text_sha256 = hashlib.sha256(
        target_text.encode(
            "utf-8"
        )
    ).hexdigest()

    metadata = (
        record[
            "source_metadata"
        ]
        or {}
    )

    speaker = (
        metadata.get(
            "speaker"
        )
        or {}
    )

    base_speaker = (
        speaker.get(
            "base"
        )
        or {}
    )

    effective_speaker = (
        speaker.get(
            "effective"
        )
        or {}
    )

    return {
        "document_id":
            document_id,

        "role":
            role,

        "char_start":
            char_start,

        "char_end":
            char_end,

        "char_count":
            (
                char_end
                -
                char_start
            ),

        "token_count":
            nb11_token_count(
                target_text
            ),

        "target_text_sha256":
            target_text_sha256,

        "source_id":
            record[
                "source_id"
            ],

        "locator":
            record[
                "locator"
            ],

        "attribution":
            record[
                "attribution"
            ],

        "speaker_state": {
            "base_status":
                base_speaker.get(
                    "status"
                ),

            "base_raw":
                base_speaker.get(
                    "raw"
                ),

            "effective_label":
                effective_speaker.get(
                    "label"
                ),

            "effective_canonical":
                effective_speaker.get(
                    "canonical"
                ),

            "effective_primary":
                effective_speaker.get(
                    "primary_dr_roudi_corpus"
                ),
        },
    }


def target_part_identity(
    part,
):

    return {
        "document_id":
            part[
                "document_id"
            ],

        "role":
            part[
                "role"
            ],

        "char_start":
            part[
                "char_start"
            ],

        "char_end":
            part[
                "char_end"
            ],
    }


def build_target_evidence_id(
    task_draft_id,
    method,
    parts,
):

    payload = {
        "schema_version":
            TARGET_EVIDENCE_SCHEMA_VERSION,

        "task_draft_id":
            task_draft_id,

        "method":
            method,

        "parts":
            [
                target_part_identity(
                    part
                )
                for part
                in parts
            ],
    }

    digest = hashlib.sha256(
        stable_json_key(
            payload
        ).encode(
            "utf-8"
        )
    ).hexdigest()

    return (
        "target-evidence:"
        + digest[
            :32
        ]
    )


def stable_target_tie_hash(
    payload,
):

    return hashlib.sha256(
        stable_json_key(
            payload
        ).encode(
            "utf-8"
        )
    ).hexdigest()


# ============================================================
# 2) Transcript speaker / authority signatures
# ============================================================

def transcript_speaker_signature(
    transcript_record,
):

    metadata = (
        transcript_record.get(
            "source_metadata"
        )
        or {}
    )

    speaker = (
        metadata.get(
            "speaker"
        )
        or {}
    )

    base = (
        speaker.get(
            "base"
        )
        or {}
    )

    effective = (
        speaker.get(
            "effective"
        )
        or {}
    )

    return (
        effective.get(
            "label"
        ),
        effective.get(
            "canonical"
        ),
        effective.get(
            "primary_dr_roudi_corpus"
        ),
        base.get(
            "status"
        ),
    )


def transcript_primary_state(
    transcript_record,
):

    metadata = (
        transcript_record.get(
            "source_metadata"
        )
        or {}
    )

    speaker = (
        metadata.get(
            "speaker"
        )
        or {}
    )

    effective = (
        speaker.get(
            "effective"
        )
        or {}
    )

    return effective.get(
        "primary_dr_roudi_corpus"
    )


# ============================================================
# 3) Deterministic local Website target
#
# Minimum multi-sentence window:
# - 2 adjacent sentences when available;
# - otherwise the only sentence.
#
# Among windows choose maximum token payload.
# Tie-break by stable structural hash.
# ============================================================

def select_website_local_target(
    document_id,
):

    record = (
        corpus_target_index[
            document_id
        ]
    )

    text = (
        record[
            "text"
        ]
    )

    sentence_spans = (
        nb11_sentence_spans_with_offsets(
            text
        )
    )

    assert sentence_spans

    window_size = min(
        2,
        len(
            sentence_spans
        ),
    )

    candidates = []

    for index in range(
        0,
        len(
            sentence_spans
        )
        - window_size
        + 1,
    ):

        left = (
            sentence_spans[
                index
            ]
        )

        right = (
            sentence_spans[
                index
                +
                window_size
                -
                1
            ]
        )

        char_start = (
            left[
                "start"
            ]
        )

        char_end = (
            right[
                "end"
            ]
        )

        target_text = text[
            char_start:
            char_end
        ]

        token_count = (
            nb11_token_count(
                target_text
            )
        )

        tie_hash = (
            stable_target_tie_hash(
                {
                    "document_id":
                        document_id,

                    "char_start":
                        char_start,

                    "char_end":
                        char_end,
                }
            )
        )

        candidates.append(
            {
                "char_start":
                    char_start,

                "char_end":
                    char_end,

                "token_count":
                    token_count,

                "tie_hash":
                    tie_hash,
            }
        )

    winner = sorted(
        candidates,
        key=lambda item: (
            -item[
                "token_count"
            ],
            item[
                "tie_hash"
            ],
        ),
    )[0]

    return [
        target_part_from_document(
            document_id=(
                document_id
            ),
            char_start=(
                winner[
                    "char_start"
                ]
            ),
            char_end=(
                winner[
                    "char_end"
                ]
            ),
            role=(
                "LOCAL_SENTENCE_WINDOW"
            ),
        )
    ]


# ============================================================
# 4) Deterministic Book long-section target
#
# Use the minimum genuine multi-record window: two adjacent
# records. Choose maximum combined token payload.
# ============================================================

def select_book_long_section_target(
    document_ids,
):

    ordered = sorted(
        document_ids,
        key=lambda document_id: (
            (
                corpus_target_index[
                    document_id
                ][
                    "source_metadata"
                ]
                or {}
            ).get(
                "paragraph_number"
            )
            if (
                (
                    corpus_target_index[
                        document_id
                    ][
                        "source_metadata"
                    ]
                    or {}
                ).get(
                    "paragraph_number"
                )
                is not None
            )
            else 10**12
        ),
    )

    assert (
        len(
            ordered
        )
        >= 2
    )

    candidates = []

    for index in range(
        len(
            ordered
        )
        - 1
    ):

        pair = [
            ordered[
                index
            ],
            ordered[
                index
                + 1
            ],
        ]

        combined_text = "\n".join(
            corpus_target_index[
                document_id
            ][
                "text"
            ]
            for document_id
            in pair
        )

        token_count = (
            nb11_token_count(
                combined_text
            )
        )

        tie_hash = (
            stable_target_tie_hash(
                {
                    "document_ids":
                        pair
                }
            )
        )

        candidates.append(
            {
                "document_ids":
                    pair,

                "token_count":
                    token_count,

                "tie_hash":
                    tie_hash,
            }
        )

    winner = sorted(
        candidates,
        key=lambda item: (
            -item[
                "token_count"
            ],
            item[
                "tie_hash"
            ],
        ),
    )[0]

    return [
        target_part_from_document(
            document_id=(
                winner[
                    "document_ids"
                ][
                    0
                ]
            ),
            role=(
                "MULTI_RECORD_WINDOW_LEFT"
            ),
        ),

        target_part_from_document(
            document_id=(
                winner[
                    "document_ids"
                ][
                    1
                ]
            ),
            role=(
                "MULTI_RECORD_WINDOW_RIGHT"
            ),
        ),
    ]


# ============================================================
# 5) Deterministic local transcript utterance
#
# Choose maximum source-utterance token payload.
# ============================================================

def select_transcript_local_utterance(
    physical_file_id,
):

    records = (
        transcript_records_by_physical_file[
            physical_file_id
        ]
    )

    assert records

    candidates = []

    for record in records:

        token_count = (
            nb11_token_count(
                record[
                    "text"
                ]
            )
        )

        tie_hash = (
            stable_target_tie_hash(
                {
                    "document_id":
                        record[
                            "document_id"
                        ]
                }
            )
        )

        candidates.append(
            {
                "document_id":
                    record[
                        "document_id"
                    ],

                "token_count":
                    token_count,

                "tie_hash":
                    tie_hash,
            }
        )

    winner = sorted(
        candidates,
        key=lambda item: (
            -item[
                "token_count"
            ],
            item[
                "tie_hash"
            ],
        ),
    )[0]

    return [
        target_part_from_document(
            document_id=(
                winner[
                    "document_id"
                ]
            ),
            role=(
                "LOCAL_UTTERANCE"
            ),
        )
    ]


# ============================================================
# 6) Attribution-aware local transcript target
#
# Find two records with different explicit attribution
# signatures.
#
# Priority:
# 1. minimum utterance-index distance;
# 2. maximum combined token payload;
# 3. stable structural hash.
#
# No speaker inference.
# ============================================================

def select_transcript_attribution_pair(
    physical_file_id,
):

    records = (
        transcript_records_by_physical_file[
            physical_file_id
        ]
    )

    assert (
        len(
            records
        )
        >= 2
    )

    candidates = []

    for left_index in range(
        len(
            records
        )
    ):

        left = (
            records[
                left_index
            ]
        )

        left_signature = (
            transcript_speaker_signature(
                left
            )
        )

        for right_index in range(
            left_index + 1,
            len(
                records
            ),
        ):

            right = (
                records[
                    right_index
                ]
            )

            right_signature = (
                transcript_speaker_signature(
                    right
                )
            )

            if (
                left_signature
                ==
                right_signature
            ):
                continue

            left_utterance_index = (
                left.get(
                    "utterance_index"
                )
            )

            right_utterance_index = (
                right.get(
                    "utterance_index"
                )
            )

            if (
                left_utterance_index
                is not None
                and
                right_utterance_index
                is not None
            ):

                distance = (
                    right_utterance_index
                    -
                    left_utterance_index
                )

            else:

                distance = (
                    right_index
                    -
                    left_index
                )

            combined_tokens = (
                nb11_token_count(
                    left[
                        "text"
                    ]
                )
                +
                nb11_token_count(
                    right[
                        "text"
                    ]
                )
            )

            tie_hash = (
                stable_target_tie_hash(
                    {
                        "left_document_id":
                            left[
                                "document_id"
                            ],

                        "right_document_id":
                            right[
                                "document_id"
                            ],
                    }
                )
            )

            candidates.append(
                {
                    "left_document_id":
                        left[
                            "document_id"
                        ],

                    "right_document_id":
                        right[
                            "document_id"
                        ],

                    "distance":
                        distance,

                    "combined_tokens":
                        combined_tokens,

                    "tie_hash":
                        tie_hash,

                    "left_signature":
                        left_signature,

                    "right_signature":
                        right_signature,
                }
            )

    assert candidates, (
        "No explicit attribution contrast found in "
        f"{physical_file_id}"
    )

    winner = sorted(
        candidates,
        key=lambda item: (
            item[
                "distance"
            ],
            -item[
                "combined_tokens"
            ],
            item[
                "tie_hash"
            ],
        ),
    )[0]

    return [
        target_part_from_document(
            document_id=(
                winner[
                    "left_document_id"
                ]
            ),
            role=(
                "ATTRIBUTION_CONTRAST_LEFT"
            ),
        ),

        target_part_from_document(
            document_id=(
                winner[
                    "right_document_id"
                ]
            ),
            role=(
                "ATTRIBUTION_CONTRAST_RIGHT"
            ),
        ),
    ]


# ============================================================
# 7) Authority-contrast transcript target
#
# Same locality rule, but contrast must be in the explicit
# effective_primary state.
# ============================================================

def select_transcript_authority_pair(
    physical_file_id,
):

    records = (
        transcript_records_by_physical_file[
            physical_file_id
        ]
    )

    assert (
        len(
            records
        )
        >= 2
    )

    candidates = []

    for left_index in range(
        len(
            records
        )
    ):

        left = (
            records[
                left_index
            ]
        )

        left_state = (
            transcript_primary_state(
                left
            )
        )

        for right_index in range(
            left_index + 1,
            len(
                records
            ),
        ):

            right = (
                records[
                    right_index
                ]
            )

            right_state = (
                transcript_primary_state(
                    right
                )
            )

            if (
                left_state
                ==
                right_state
            ):
                continue

            left_utterance_index = (
                left.get(
                    "utterance_index"
                )
            )

            right_utterance_index = (
                right.get(
                    "utterance_index"
                )
            )

            if (
                left_utterance_index
                is not None
                and
                right_utterance_index
                is not None
            ):

                distance = (
                    right_utterance_index
                    -
                    left_utterance_index
                )

            else:

                distance = (
                    right_index
                    -
                    left_index
                )

            combined_tokens = (
                nb11_token_count(
                    left[
                        "text"
                    ]
                )
                +
                nb11_token_count(
                    right[
                        "text"
                    ]
                )
            )

            tie_hash = (
                stable_target_tie_hash(
                    {
                        "left_document_id":
                            left[
                                "document_id"
                            ],

                        "right_document_id":
                            right[
                                "document_id"
                            ],
                    }
                )
            )

            candidates.append(
                {
                    "left_document_id":
                        left[
                            "document_id"
                        ],

                    "right_document_id":
                        right[
                            "document_id"
                        ],

                    "left_state":
                        left_state,

                    "right_state":
                        right_state,

                    "distance":
                        distance,

                    "combined_tokens":
                        combined_tokens,

                    "tie_hash":
                        tie_hash,
                }
            )

    assert candidates, (
        "No explicit primary-state contrast found in "
        f"{physical_file_id}"
    )

    winner = sorted(
        candidates,
        key=lambda item: (
            item[
                "distance"
            ],
            -item[
                "combined_tokens"
            ],
            item[
                "tie_hash"
            ],
        ),
    )[0]

    return [
        target_part_from_document(
            document_id=(
                winner[
                    "left_document_id"
                ]
            ),
            role=(
                "AUTHORITY_CONTRAST_LEFT"
            ),
        ),

        target_part_from_document(
            document_id=(
                winner[
                    "right_document_id"
                ]
            ),
            role=(
                "AUTHORITY_CONTRAST_RIGHT"
            ),
        ),
    ]


# ============================================================
# 8) Main method executor
# ============================================================

def select_target_evidence(
    plan,
    case_record,
):

    method = (
        plan[
            "evidence_target_method"
        ]
    )

    candidate = (
        case_record[
            "candidate_snapshot"
        ]
    )


    # --------------------------------------------------------
    # Exact full-document target
    # --------------------------------------------------------

    if method == "DOCUMENT_LEVEL_TARGET":

        parts = [
            target_part_from_document(
                document_id=(
                    candidate[
                        "document_id"
                    ]
                ),
                role=(
                    "SOURCE_DOCUMENT"
                ),
            )
        ]

        selection_basis = (
            "EXACT_CASE_SOURCE_DOCUMENT"
        )


    # --------------------------------------------------------
    # Long Website local target
    # --------------------------------------------------------

    elif (
        method
        ==
        "DETERMINISTIC_LOCAL_EVIDENCE_TARGET_REQUIRED"
    ):

        parts = (
            select_website_local_target(
                candidate[
                    "document_id"
                ]
            )
        )

        selection_basis = (
            "MAX_TOKEN_ADJACENT_SENTENCE_WINDOW"
        )


    # --------------------------------------------------------
    # Exact protected Book region
    # --------------------------------------------------------

    elif method in {
        "USE_EXACT_PROTECTED_REGION",
        "USE_EXACT_PROTECTED_TABLE_REGION",
    }:

        parts = [
            target_part_from_document(
                document_id=(
                    document_id
                ),
                role=(
                    "PROTECTED_REGION_MEMBER"
                ),
            )

            for document_id
            in candidate[
                "member_document_ids"
            ]
        ]

        selection_basis = (
            "EXACT_FROZEN_PROTECTED_REGION"
        )


    # --------------------------------------------------------
    # Long Book section — deterministic 2-record window
    # --------------------------------------------------------

    elif (
        method
        ==
        "DETERMINISTIC_LOCAL_OR_MULTI_RECORD_TARGET_REQUIRED"
    ):

        parts = (
            select_book_long_section_target(
                candidate[
                    "document_ids"
                ]
            )
        )

        selection_basis = (
            "MAX_TOKEN_MINIMAL_TWO_RECORD_CONTIGUOUS_WINDOW"
        )


    # --------------------------------------------------------
    # Exact table run
    # --------------------------------------------------------

    elif method == "USE_EXACT_TABLE_RUN":

        parts = [
            target_part_from_document(
                document_id=(
                    document_id
                ),
                role=(
                    "TABLE_RUN_MEMBER"
                ),
            )

            for document_id
            in candidate[
                "document_ids"
            ]
        ]

        selection_basis = (
            "EXACT_FROZEN_CONTIGUOUS_TABLE_RUN"
        )


    # --------------------------------------------------------
    # Explicit exercise/self-assessment Book region
    # --------------------------------------------------------

    elif method == "USE_EXPLICIT_BOOK_REGION":

        parts = [
            target_part_from_document(
                document_id=(
                    candidate[
                        "document_id"
                    ]
                ),
                role=(
                    "EXPLICIT_BOOK_REGION"
                ),
            )
        ]

        selection_basis = (
            "EXACT_EXPLICIT_BOOK_REGION"
        )


    # --------------------------------------------------------
    # Parent + supplemental
    # --------------------------------------------------------

    elif (
        method
        ==
        "USE_PARENT_PLUS_SUPPLEMENTAL_RELATION"
    ):

        parts = [
            target_part_from_document(
                document_id=(
                    candidate[
                        "parent_paragraph_id"
                    ]
                ),
                role=(
                    "PARENT_PARAGRAPH"
                ),
            ),

            target_part_from_document(
                document_id=(
                    candidate[
                        "document_id"
                    ]
                ),
                role=(
                    "SUPPLEMENTAL_RECORD"
                ),
            ),
        ]

        selection_basis = (
            "EXACT_PARENT_SUPPLEMENTAL_RELATION"
        )


    # --------------------------------------------------------
    # Exact Book record + heading-path context
    # --------------------------------------------------------

    elif method == "USE_RECORD_PLUS_HEADING_PATH":

        parts = [
            target_part_from_document(
                document_id=(
                    candidate[
                        "document_id"
                    ]
                ),
                role=(
                    "HIERARCHICAL_BOOK_RECORD"
                ),
            )
        ]

        selection_basis = (
            "EXACT_BOOK_RECORD_WITH_HEADING_PATH"
        )


    # --------------------------------------------------------
    # Local transcript utterance
    # --------------------------------------------------------

    elif (
        method
        ==
        "DETERMINISTIC_LOCAL_UTTERANCE_TARGET_REQUIRED"
    ):

        parts = (
            select_transcript_local_utterance(
                candidate[
                    "physical_file_id"
                ]
            )
        )

        selection_basis = (
            "MAX_TOKEN_UTTERANCE_IN_PHYSICAL_FILE"
        )


    # --------------------------------------------------------
    # Attribution-aware transcript pair
    # --------------------------------------------------------

    elif (
        method
        ==
        "DETERMINISTIC_ATTRIBUTION_AWARE_LOCAL_TARGET_REQUIRED"
    ):

        parts = (
            select_transcript_attribution_pair(
                candidate[
                    "physical_file_id"
                ]
            )
        )

        selection_basis = (
            "NEAREST_EXPLICIT_ATTRIBUTION_CONTRAST_PAIR"
        )


    # --------------------------------------------------------
    # Exact primary -> nonprimary adjacent pair
    # --------------------------------------------------------

    elif method == "USE_EXACT_ADJACENT_PAIR":

        parts = [
            target_part_from_document(
                document_id=(
                    candidate[
                        "left_document_id"
                    ]
                ),
                role=(
                    "ADJACENT_PAIR_LEFT"
                ),
            ),

            target_part_from_document(
                document_id=(
                    candidate[
                        "right_document_id"
                    ]
                ),
                role=(
                    "ADJACENT_PAIR_RIGHT"
                ),
            ),
        ]

        selection_basis = (
            "EXACT_VERIFIED_ADJACENT_UTTERANCE_PAIR"
        )


    # --------------------------------------------------------
    # Mixed-authority local pair
    # --------------------------------------------------------

    elif (
        method
        ==
        "DETERMINISTIC_AUTHORITY_CONTRAST_TARGET_REQUIRED"
    ):

        parts = (
            select_transcript_authority_pair(
                candidate[
                    "physical_file_id"
                ]
            )
        )

        selection_basis = (
            "NEAREST_EXPLICIT_PRIMARY_STATE_CONTRAST_PAIR"
        )


    # --------------------------------------------------------
    # Exact long utterance
    # --------------------------------------------------------

    elif method == "USE_EXACT_SOURCE_UTTERANCE":

        parts = [
            target_part_from_document(
                document_id=(
                    candidate[
                        "document_id"
                    ]
                ),
                role=(
                    "SOURCE_UTTERANCE"
                ),
            )
        ]

        selection_basis = (
            "EXACT_SOURCE_UTTERANCE"
        )


    # --------------------------------------------------------
    # Exact fallback sentence
    # --------------------------------------------------------

    elif method == "USE_EXACT_SOURCE_SENTENCE":

        parts = [
            target_part_from_document(
                document_id=(
                    candidate[
                        "document_id"
                    ]
                ),
                char_start=(
                    candidate[
                        "char_start"
                    ]
                ),
                char_end=(
                    candidate[
                        "char_end"
                    ]
                ),
                role=(
                    "SOURCE_SENTENCE"
                ),
            )
        ]

        selection_basis = (
            "EXACT_FROZEN_SOURCE_SENTENCE_SPAN"
        )


    else:

        raise AssertionError(
            "Executable target method not implemented: "
            f"{method}"
        )


    target_evidence_id = (
        build_target_evidence_id(
            task_draft_id=(
                plan[
                    "task_draft_id"
                ]
            ),
            method=(
                method
            ),
            parts=(
                parts
            ),
        )
    )


    return {
        "target_evidence_id":
            target_evidence_id,

        "target_schema_version":
            TARGET_EVIDENCE_SCHEMA_VERSION,

        "task_draft_id":
            plan[
                "task_draft_id"
            ],

        "golden_case_id":
            plan[
                "golden_case_id"
            ],

        "source_type":
            plan[
                "source_type"
            ],

        "stratum":
            plan[
                "stratum"
            ],

        "evidence_target_method":
            method,

        "target_selection_basis":
            selection_basis,

        "target_parts":
            parts,

        "target_document_ids":
            [
                part[
                    "document_id"
                ]
                for part
                in parts
            ],

        "target_locator_refs":
            [
                {
                    "document_id":
                        part[
                            "document_id"
                        ],

                    "locator":
                        part[
                            "locator"
                        ],
                }
                for part
                in parts
            ],

        "target_text_spans":
            [
                {
                    "document_id":
                        part[
                            "document_id"
                        ],

                    "role":
                        part[
                            "role"
                        ],

                    "char_start":
                        part[
                            "char_start"
                        ],

                    "char_end":
                        part[
                            "char_end"
                        ],

                    "target_text_sha256":
                        part[
                            "target_text_sha256"
                        ],
                }
                for part
                in parts
            ],

        "target_attribution_state":
            [
                {
                    "document_id":
                        part[
                            "document_id"
                        ],

                    "attribution":
                        part[
                            "attribution"
                        ],

                    "speaker_state":
                        part[
                            "speaker_state"
                        ],
                }
                for part
                in parts
            ],

        "target_authority_state": {
            "task_authority_expectation":
                plan[
                    "authority_expectation"
                ],

            "authority_inference_performed":
                False,
        },

        "target_total_tokens":
            sum(
                part[
                    "token_count"
                ]
                for part
                in parts
            ),

        "question_generation_status":
            "NOT_STARTED",

        "expected_answer_generation_status":
            "NOT_STARTED",

        "citation_target_status":
            "TARGET_LOCATORS_CAPTURED_NOT_YET_JUDGED",
    }


# ============================================================
# 9) Execute selection for all 572 ready cases
# ============================================================

target_evidence_records = []

target_evidence_by_task_draft_id = {}

target_evidence_by_case_id = {}


for plan in (
    task_draft_plan_rows
):

    if (
        plan[
            "construction_eligibility"
        ]
        !=
        "READY_FOR_TARGET_EVIDENCE_SELECTION"
    ):
        continue

    case_record = (
        golden_case_records_by_id[
            plan[
                "golden_case_id"
            ]
        ]
    )

    target_record = (
        select_target_evidence(
            plan=(
                plan
            ),
            case_record=(
                case_record
            ),
        )
    )

    target_evidence_records.append(
        target_record
    )

    target_evidence_by_task_draft_id[
        plan[
            "task_draft_id"
        ]
    ] = (
        target_record
    )

    target_evidence_by_case_id[
        plan[
            "golden_case_id"
        ]
    ] = (
        target_record
    )


# ============================================================
# 10) Independent deterministic recomputation
# ============================================================

target_recomputation_failures = []


for plan in (
    task_draft_plan_rows
):

    if (
        plan[
            "construction_eligibility"
        ]
        !=
        "READY_FOR_TARGET_EVIDENCE_SELECTION"
    ):
        continue

    case_record = (
        golden_case_records_by_id[
            plan[
                "golden_case_id"
            ]
        ]
    )

    first = (
        target_evidence_by_task_draft_id[
            plan[
                "task_draft_id"
            ]
        ]
    )

    second = (
        select_target_evidence(
            plan=(
                plan
            ),
            case_record=(
                case_record
            ),
        )
    )

    if (
        first[
            "target_evidence_id"
        ]
        !=
        second[
            "target_evidence_id"
        ]

        or

        first[
            "target_text_spans"
        ]
        !=
        second[
            "target_text_spans"
        ]
    ):

        target_recomputation_failures.append(
            plan[
                "task_draft_id"
            ]
        )


# ============================================================
# 11) Independent integrity audits
# ============================================================

target_subset_failures = []

target_span_failures = []

target_empty_text_failures = []

target_duplicate_id_failures = []

target_method_semantic_failures = []


target_id_counts = Counter(
    record[
        "target_evidence_id"
    ]
    for record
    in target_evidence_records
)


target_duplicate_id_failures = [
    target_id
    for target_id, count
    in target_id_counts.items()
    if count > 1
]


for target_record in (
    target_evidence_records
):

    case_record = (
        golden_case_records_by_id[
            target_record[
                "golden_case_id"
            ]
        ]
    )

    case_evidence_ids = set(
        case_record[
            "evidence_document_ids"
        ]
    )

    target_document_ids = (
        target_record[
            "target_document_ids"
        ]
    )


    # --------------------------------------------------------
    # Target must be entirely inside validated case evidence.
    # --------------------------------------------------------

    if not set(
        target_document_ids
    ).issubset(
        case_evidence_ids
    ):

        target_subset_failures.append(
            target_record[
                "target_evidence_id"
            ]
        )


    # --------------------------------------------------------
    # Span integrity
    # --------------------------------------------------------

    for span in (
        target_record[
            "target_text_spans"
        ]
    ):

        document_id = (
            span[
                "document_id"
            ]
        )

        text = (
            corpus_target_index[
                document_id
            ][
                "text"
            ]
        )

        char_start = (
            span[
                "char_start"
            ]
        )

        char_end = (
            span[
                "char_end"
            ]
        )

        valid = (
            0
            <= char_start
            < char_end
            <= len(
                text
            )
        )

        if not valid:

            target_span_failures.append(
                {
                    "target_evidence_id":
                        target_record[
                            "target_evidence_id"
                        ],

                    "document_id":
                        document_id,
                }
            )

            continue


        target_text = text[
            char_start:
            char_end
        ]

        if not target_text.strip():

            target_empty_text_failures.append(
                {
                    "target_evidence_id":
                        target_record[
                            "target_evidence_id"
                        ],

                    "document_id":
                        document_id,
                }
            )


        recomputed_sha = hashlib.sha256(
            target_text.encode(
                "utf-8"
            )
        ).hexdigest()

        if (
            recomputed_sha
            !=
            span[
                "target_text_sha256"
            ]
        ):

            target_span_failures.append(
                {
                    "target_evidence_id":
                        target_record[
                            "target_evidence_id"
                        ],

                    "document_id":
                        document_id,

                    "reason":
                        "TEXT_SHA_MISMATCH",
                }
            )


    # --------------------------------------------------------
    # Method-specific semantic/structural assertions
    # --------------------------------------------------------

    method = (
        target_record[
            "evidence_target_method"
        ]
    )

    parts = (
        target_record[
            "target_parts"
        ]
    )

    candidate = (
        case_record[
            "candidate_snapshot"
        ]
    )


    if method == "DOCUMENT_LEVEL_TARGET":

        valid_method = (
            len(
                parts
            )
            == 1

            and

            parts[
                0
            ][
                "document_id"
            ]
            ==
            candidate[
                "document_id"
            ]
        )


    elif (
        method
        ==
        "DETERMINISTIC_LOCAL_EVIDENCE_TARGET_REQUIRED"
    ):

        valid_method = (
            len(
                parts
            )
            == 1

            and

            parts[
                0
            ][
                "document_id"
            ]
            ==
            candidate[
                "document_id"
            ]

            and

            parts[
                0
            ][
                "char_count"
            ]
            > 0
        )


    elif method in {
        "USE_EXACT_PROTECTED_REGION",
        "USE_EXACT_PROTECTED_TABLE_REGION",
    }:

        valid_method = (
            target_document_ids
            ==
            candidate[
                "member_document_ids"
            ]
        )


    elif (
        method
        ==
        "DETERMINISTIC_LOCAL_OR_MULTI_RECORD_TARGET_REQUIRED"
    ):

        candidate_ids = (
            candidate[
                "document_ids"
            ]
        )

        valid_method = (
            len(
                parts
            )
            == 2

            and

            all(
                document_id
                in candidate_ids

                for document_id
                in target_document_ids
            )
        )


    elif method == "USE_EXACT_TABLE_RUN":

        valid_method = (
            target_document_ids
            ==
            candidate[
                "document_ids"
            ]
        )


    elif method == "USE_EXPLICIT_BOOK_REGION":

        valid_method = (
            len(
                parts
            )
            == 1

            and

            parts[
                0
            ][
                "document_id"
            ]
            ==
            candidate[
                "document_id"
            ]
        )


    elif (
        method
        ==
        "USE_PARENT_PLUS_SUPPLEMENTAL_RELATION"
    ):

        valid_method = (
            len(
                parts
            )
            == 2

            and

            parts[
                0
            ][
                "document_id"
            ]
            ==
            candidate[
                "parent_paragraph_id"
            ]

            and

            parts[
                1
            ][
                "document_id"
            ]
            ==
            candidate[
                "document_id"
            ]
        )


    elif method == "USE_RECORD_PLUS_HEADING_PATH":

        valid_method = (
            len(
                parts
            )
            == 1

            and

            parts[
                0
            ][
                "document_id"
            ]
            ==
            candidate[
                "document_id"
            ]

            and

            bool(
                candidate.get(
                    "heading_path"
                )
            )
        )


    elif (
        method
        ==
        "DETERMINISTIC_LOCAL_UTTERANCE_TARGET_REQUIRED"
    ):

        valid_method = (
            len(
                parts
            )
            == 1

            and

            parts[
                0
            ][
                "document_id"
            ]
            in
            case_record[
                "evidence_document_ids"
            ]
        )


    elif (
        method
        ==
        "DETERMINISTIC_ATTRIBUTION_AWARE_LOCAL_TARGET_REQUIRED"
    ):

        left_signature = (
            (
                parts[
                    0
                ][
                    "speaker_state"
                ][
                    "effective_label"
                ],
                parts[
                    0
                ][
                    "speaker_state"
                ][
                    "effective_canonical"
                ],
                parts[
                    0
                ][
                    "speaker_state"
                ][
                    "effective_primary"
                ],
                parts[
                    0
                ][
                    "speaker_state"
                ][
                    "base_status"
                ],
            )
        )

        right_signature = (
            (
                parts[
                    1
                ][
                    "speaker_state"
                ][
                    "effective_label"
                ],
                parts[
                    1
                ][
                    "speaker_state"
                ][
                    "effective_canonical"
                ],
                parts[
                    1
                ][
                    "speaker_state"
                ][
                    "effective_primary"
                ],
                parts[
                    1
                ][
                    "speaker_state"
                ][
                    "base_status"
                ],
            )
        )

        valid_method = (
            len(
                parts
            )
            == 2

            and

            left_signature
            !=
            right_signature
        )


    elif method == "USE_EXACT_ADJACENT_PAIR":

        valid_method = (
            len(
                parts
            )
            == 2

            and

            parts[
                0
            ][
                "document_id"
            ]
            ==
            candidate[
                "left_document_id"
            ]

            and

            parts[
                1
            ][
                "document_id"
            ]
            ==
            candidate[
                "right_document_id"
            ]
        )


    elif (
        method
        ==
        "DETERMINISTIC_AUTHORITY_CONTRAST_TARGET_REQUIRED"
    ):

        left_primary = (
            parts[
                0
            ][
                "speaker_state"
            ][
                "effective_primary"
            ]
        )

        right_primary = (
            parts[
                1
            ][
                "speaker_state"
            ][
                "effective_primary"
            ]
        )

        valid_method = (
            len(
                parts
            )
            == 2

            and

            left_primary
            !=
            right_primary
        )


    elif method == "USE_EXACT_SOURCE_UTTERANCE":

        valid_method = (
            len(
                parts
            )
            == 1

            and

            parts[
                0
            ][
                "document_id"
            ]
            ==
            candidate[
                "document_id"
            ]

            and

            parts[
                0
            ][
                "char_start"
            ]
            == 0

            and

            parts[
                0
            ][
                "char_end"
            ]
            ==
            len(
                corpus_target_index[
                    candidate[
                        "document_id"
                    ]
                ][
                    "text"
                ]
            )
        )


    elif method == "USE_EXACT_SOURCE_SENTENCE":

        valid_method = (
            len(
                parts
            )
            == 1

            and

            parts[
                0
            ][
                "document_id"
            ]
            ==
            candidate[
                "document_id"
            ]

            and

            parts[
                0
            ][
                "char_start"
            ]
            ==
            candidate[
                "char_start"
            ]

            and

            parts[
                0
            ][
                "char_end"
            ]
            ==
            candidate[
                "char_end"
            ]
        )


    else:

        valid_method = False


    if not valid_method:

        target_method_semantic_failures.append(
            {
                "target_evidence_id":
                    target_record[
                        "target_evidence_id"
                    ],

                "method":
                    method,

                "golden_case_id":
                    target_record[
                        "golden_case_id"
                    ],
            }
        )


# ============================================================
# 12) Method accounting
# ============================================================

selected_target_method_counts = Counter(
    record[
        "evidence_target_method"
    ]
    for record
    in target_evidence_records
)


expected_ready_method_counts = Counter(
    plan[
        "evidence_target_method"
    ]
    for plan
    in task_draft_plan_rows
    if (
        plan[
            "construction_eligibility"
        ]
        ==
        "READY_FOR_TARGET_EVIDENCE_SELECTION"
    )
)


method_count_mismatches = {
    method: {
        "expected":
            expected_ready_method_counts[
                method
            ],

        "actual":
            selected_target_method_counts[
                method
            ],
    }

    for method
    in set(
        expected_ready_method_counts
    )
    |
    set(
        selected_target_method_counts
    )

    if (
        expected_ready_method_counts[
            method
        ]
        !=
        selected_target_method_counts[
            method
        ]
    )
}


# ============================================================
# 13) Review leakage audit
# ============================================================

blocked_case_ids = {
    plan[
        "golden_case_id"
    ]

    for plan
    in task_draft_plan_rows

    if (
        plan[
            "construction_eligibility"
        ]
        ==
        "BLOCKED_PENDING_HUMAN_CONFIRMATION"
    )
}


review_target_leakage = [
    record[
        "golden_case_id"
    ]

    for record
    in target_evidence_records

    if (
        record[
            "golden_case_id"
        ]
        in blocked_case_ids
    )
]


# ============================================================
# 14) Target Evidence Registry draft
# ============================================================

TARGET_EVIDENCE_REGISTRY_V1_DRAFT = {
    "version":
        "target_evidence_registry_v1_draft",

    "status":
        "SELECTED_AND_VALIDATED_IN_MEMORY_NOT_FROZEN",

    "target_schema_version":
        TARGET_EVIDENCE_SCHEMA_VERSION,

    "ready_case_count":
        len(
            target_evidence_records
        ),

    "blocked_review_case_count":
        len(
            blocked_case_ids
        ),

    "selection_method":
        "METHOD_SPECIFIC_DETERMINISTIC_SELECTION",

    "random_sampling":
        False,

    "random_seed":
        None,

    "target_before_question":
        True,

    "question_generation_status":
        "NOT_STARTED",

    "expected_answer_generation_status":
        "NOT_STARTED",

    "retrieval_execution_status":
        "NOT_STARTED",

    "artifact_write_status":
        "NONE",
}


# ============================================================
# 15) Validation gates
# ============================================================

exact_572_targets_gate = (
    len(
        target_evidence_records
    )
    == 572
)


one_target_per_ready_case_gate = (
    len(
        target_evidence_by_case_id
    )
    == 572

    and

    len(
        target_evidence_by_task_draft_id
    )
    == 572
)


target_ids_unique_gate = (
    len(
        target_duplicate_id_failures
    )
    == 0
)


target_recomputation_gate = (
    len(
        target_recomputation_failures
    )
    == 0
)


target_subset_gate = (
    len(
        target_subset_failures
    )
    == 0
)


target_span_integrity_gate = (
    len(
        target_span_failures
    )
    == 0

    and

    len(
        target_empty_text_failures
    )
    == 0
)


method_semantics_gate = (
    len(
        target_method_semantic_failures
    )
    == 0
)


method_counts_match_gate = (
    len(
        method_count_mismatches
    )
    == 0
)


exact_152_review_blocked_gate = (
    len(
        blocked_case_ids
    )
    == 152
)


no_review_target_leakage_gate = (
    len(
        review_target_leakage
    )
    == 0
)


no_randomness_gate = (
    TARGET_EVIDENCE_REGISTRY_V1_DRAFT[
        "random_sampling"
    ]
    is False

    and

    TARGET_EVIDENCE_REGISTRY_V1_DRAFT[
        "random_seed"
    ]
    is None
)


question_generation_not_started_gate = (
    TARGET_EVIDENCE_REGISTRY_V1_DRAFT[
        "question_generation_status"
    ]
    == "NOT_STARTED"

    and

    all(
        record[
            "question_generation_status"
        ]
        == "NOT_STARTED"

        for record
        in target_evidence_records
    )
)


expected_answer_not_started_gate = (
    TARGET_EVIDENCE_REGISTRY_V1_DRAFT[
        "expected_answer_generation_status"
    ]
    == "NOT_STARTED"

    and

    all(
        record[
            "expected_answer_generation_status"
        ]
        == "NOT_STARTED"

        for record
        in target_evidence_records
    )
)


retrieval_not_started_gate = (
    TARGET_EVIDENCE_REGISTRY_V1_DRAFT[
        "retrieval_execution_status"
    ]
    == "NOT_STARTED"
)


no_artifact_write_gate = (
    TARGET_EVIDENCE_REGISTRY_V1_DRAFT[
        "artifact_write_status"
    ]
    == "NONE"
)


cell_28_gates = {
    "exact_572_target_evidence_records":
        exact_572_targets_gate,

    "exactly_one_target_per_ready_case":
        one_target_per_ready_case_gate,

    "target_evidence_ids_unique":
        target_ids_unique_gate,

    "independent_target_recomputation_matches":
        target_recomputation_gate,

    "all_targets_are_subset_of_validated_case_evidence":
        target_subset_gate,

    "all_target_text_spans_valid_and_nonempty":
        target_span_integrity_gate,

    "all_method_specific_target_semantics_pass":
        method_semantics_gate,

    "selected_target_method_counts_match_plan":
        method_counts_match_gate,

    "exact_152_review_cases_remain_without_target":
        exact_152_review_blocked_gate,

    "no_review_case_target_leakage":
        no_review_target_leakage_gate,

    "target_selection_contains_no_randomness":
        no_randomness_gate,

    "question_generation_not_started":
        question_generation_not_started_gate,

    "expected_answer_generation_not_started":
        expected_answer_not_started_gate,

    "retrieval_execution_not_started":
        retrieval_not_started_gate,

    "no_artifact_write":
        no_artifact_write_gate,
}


cell_28_pass = all(
    cell_28_gates.values()
)


# ============================================================
# 16) Update task-plan status only after successful validation
# ============================================================

if cell_28_pass:

    for plan in (
        task_draft_plan_rows
    ):

        if (
            plan[
                "construction_eligibility"
            ]
            ==
            "READY_FOR_TARGET_EVIDENCE_SELECTION"
        ):

            plan[
                "target_evidence_status"
            ] = (
                "SELECTED_AND_VALIDATED"
            )

            plan[
                "construction_eligibility"
            ] = (
                "READY_FOR_QUESTION_DRAFT"
            )


    QUESTION_CONSTRUCTION_METHOD_V1_DRAFT[
        "target_evidence_selection_status"
    ] = (
        "EXECUTED_AND_VALIDATED_IN_MEMORY"
    )


# ============================================================
# 17) Output
# ============================================================

print(
    "=== TARGET EVIDENCE SELECTION ==="
)

print(
    "Validated target records:",
    len(
        target_evidence_records
    )
)

print(
    "Review cases without target:",
    len(
        blocked_case_ids
    )
)


print(
    "\n=== TARGET METHOD COUNTS ==="
)

for method, count in sorted(
    selected_target_method_counts.items()
):

    print(
        "-",
        method,
        ":",
        count,
    )


print(
    "\n=== TARGET PART COUNT DISTRIBUTION ==="
)

target_part_count_distribution = Counter(
    len(
        record[
            "target_parts"
        ]
    )
    for record
    in target_evidence_records
)

for part_count, count in sorted(
    target_part_count_distribution.items()
):

    print(
        "- parts:",
        part_count,
        "| targets:",
        count,
    )


print(
    "\n=== REPRESENTATIVE TARGETS BY METHOD ==="
)

shown_methods = set()

for record in (
    target_evidence_records
):

    method = (
        record[
            "evidence_target_method"
        ]
    )

    if (
        method
        in shown_methods
    ):
        continue

    shown_methods.add(
        method
    )

    print(
        "\n[",
        method,
        "]",
        sep="",
    )

    print(
        json.dumps(
            {
                "target_evidence_id":
                    record[
                        "target_evidence_id"
                    ],

                "golden_case_id":
                    record[
                        "golden_case_id"
                    ],

                "source_type":
                    record[
                        "source_type"
                    ],

                "stratum":
                    record[
                        "stratum"
                    ],

                "target_selection_basis":
                    record[
                        "target_selection_basis"
                    ],

                "target_document_ids":
                    record[
                        "target_document_ids"
                    ],

                "target_text_spans":
                    record[
                        "target_text_spans"
                    ],

                "target_total_tokens":
                    record[
                        "target_total_tokens"
                    ],
            },
            ensure_ascii=False,
            sort_keys=True,
        )
    )


print(
    "\n=== TARGET VALIDATION FAILURES ==="
)

print(
    "Recomputation failures:",
    len(
        target_recomputation_failures
    )
)

print(
    "Subset failures:",
    len(
        target_subset_failures
    )
)

print(
    "Span failures:",
    len(
        target_span_failures
    )
)

print(
    "Empty target text failures:",
    len(
        target_empty_text_failures
    )
)

print(
    "Duplicate target IDs:",
    len(
        target_duplicate_id_failures
    )
)

print(
    "Method semantic failures:",
    len(
        target_method_semantic_failures
    )
)

print(
    "Method count mismatches:",
    len(
        method_count_mismatches
    )
)

print(
    "Review target leakage:",
    len(
        review_target_leakage
    )
)


print(
    "\n=== CODE CELL 28 GATES ==="
)

for name, passed in (
    cell_28_gates.items()
):

    print(
        name,
        ":",
        "PASS"
        if passed
        else "FAIL"
    )


print(
    "\nCODE CELL 28 GATE:",
    "PASS"
    if cell_28_pass
    else "FAIL"
)


assert cell_28_pass, (
    "Target Evidence Selection failed. "
    "Do not generate Golden questions until every target is "
    "deterministic, structurally valid, inside the Golden Case "
    "evidence scope, and independently reproducible."
)
```

<div style="direction: rtl; white-space: normal; line-height: 1.7;">

<h3>Representative Question-Drafting Pilot Cohort</h3>

<p>
Target Evidence برای تمام ۵۷۲ Case آماده و Validate شده است.
اما تولید هم‌زمان صدها سؤال قبل از مشاهده‌ی کیفیت واقعی Draftها ریسک ایجاد
Question Artifact، Template Bias و Groundability Failure دارد.
</p>

<p>
بنابراین قبل از Full-scale Question Generation یک Pilot کنترل‌شده ساخته می‌شود:
دقیقاً <strong>یک Case از هر یک از ۲۳ Stratum آماده</strong>.
چهار Stratum وابسته به Human Review وارد Pilot نمی‌شوند.
</p>

<p>
Pilot Case هر Stratum به‌صورت deterministic و evidence-derived انتخاب می‌شود:
Caseی که <code>target_total_tokens</code> آن به Median واقعی Target Size همان Stratum
نزدیک‌تر باشد. در صورت Tie، Stable SHA256 تصمیم‌گیر است.
</p>

<p>
برای هر Pilot Case یک <code>Drafting Packet</code> ساخته می‌شود که شامل
Task Family، Question Archetype، Evaluation Dimensions، Target Evidence Text،
Locator، Attribution/Authority State و Guardrailهای لازم است.
</p>

<p>
این Packet همان ورودی کنترل‌شده‌ی Question Drafting خواهد بود.
در این Cell هنوز Question Text یا Expected Answer تولید نمی‌شود.
</p>

</div>

```python
# Code Cell 29

# ============================================================
# Representative Question-Drafting Pilot Cohort
#
# One deterministic representative case from each of the
# 23 task-ready strata.
#
# NO question generation.
# NO expected-answer generation.
# NO retrieval execution.
# NO artifact writes.
# ============================================================


# ============================================================
# 1) Ready strata
# ============================================================

ready_stratum_rows = [
    row
    for row
    in task_design_rows
    if (
        row[
            "task_construction_status"
        ]
        ==
        "READY_FOR_TASK_CONSTRUCTION"
    )
]


assert (
    len(
        ready_stratum_rows
    )
    == 23
)


# ============================================================
# 2) Group validated Target Evidence by stratum
# ============================================================

targets_by_stratum = defaultdict(
    list
)


for target_record in (
    target_evidence_records
):

    targets_by_stratum[
        (
            target_record[
                "source_type"
            ],
            target_record[
                "stratum"
            ],
        )
    ].append(
        target_record
    )


# ============================================================
# 3) Select one representative case per ready stratum
#
# Rule:
# - empirical median target_total_tokens within stratum;
# - minimize absolute distance to median;
# - SHA256 stable tie-break.
# ============================================================

pilot_selection_rows = []


for stratum_row in (
    ready_stratum_rows
):

    source_type = (
        stratum_row[
            "source_type"
        ]
    )

    stratum = (
        stratum_row[
            "stratum"
        ]
    )

    targets = (
        targets_by_stratum[
            (
                source_type,
                stratum,
            )
        ]
    )

    assert targets


    target_token_values = [
        target[
            "target_total_tokens"
        ]
        for target
        in targets
    ]


    stratum_median_tokens = (
        empirical_nearest_rank(
            target_token_values,
            0.50,
        )
    )


    ranked_targets = []


    for target in (
        targets
    ):

        distance_to_median = abs(
            target[
                "target_total_tokens"
            ]
            -
            stratum_median_tokens
        )


        stable_tie_break = hashlib.sha256(
            stable_json_key(
                {
                    "source_type":
                        source_type,

                    "stratum":
                        stratum,

                    "target_evidence_id":
                        target[
                            "target_evidence_id"
                        ],
                }
            ).encode(
                "utf-8"
            )
        ).hexdigest()


        ranked_targets.append(
            {
                "target":
                    target,

                "distance_to_median":
                    distance_to_median,

                "stable_tie_break":
                    stable_tie_break,
            }
        )


    winner = sorted(
        ranked_targets,
        key=lambda item: (
            item[
                "distance_to_median"
            ],
            item[
                "stable_tie_break"
            ],
        ),
    )[0]


    selected_target = (
        winner[
            "target"
        ]
    )


    pilot_selection_rows.append(
        {
            "source_type":
                source_type,

            "stratum":
                stratum,

            "stratum_case_count":
                len(
                    targets
                ),

            "stratum_target_token_median":
                stratum_median_tokens,

            "selected_target_total_tokens":
                selected_target[
                    "target_total_tokens"
                ],

            "distance_to_median":
                winner[
                    "distance_to_median"
                ],

            "golden_case_id":
                selected_target[
                    "golden_case_id"
                ],

            "task_draft_id":
                selected_target[
                    "task_draft_id"
                ],

            "target_evidence_id":
                selected_target[
                    "target_evidence_id"
                ],

            "selection_tie_break_sha256":
                winner[
                    "stable_tie_break"
                ],

            "selection_basis":
                "CLOSEST_TO_EMPIRICAL_STRATUM_TARGET_TOKEN_MEDIAN",
        }
    )


# ============================================================
# 4) Stable Pilot Case ID
# ============================================================

QUESTION_DRAFTING_PILOT_VERSION = (
    "question_drafting_pilot_v1_draft"
)


def build_pilot_case_id(
    source_type,
    stratum,
    golden_case_id,
    target_evidence_id,
):

    payload = {
        "pilot_version":
            QUESTION_DRAFTING_PILOT_VERSION,

        "source_type":
            source_type,

        "stratum":
            stratum,

        "golden_case_id":
            golden_case_id,

        "target_evidence_id":
            target_evidence_id,
    }

    digest = hashlib.sha256(
        stable_json_key(
            payload
        ).encode(
            "utf-8"
        )
    ).hexdigest()

    return (
        "question-pilot:"
        + digest[
            :32
        ]
    )


# ============================================================
# 5) Build Drafting Packets
# ============================================================

question_drafting_pilot_packets = []

question_drafting_pilot_by_id = {}


for selection in (
    pilot_selection_rows
):

    case_id = (
        selection[
            "golden_case_id"
        ]
    )

    task_draft_id = (
        selection[
            "task_draft_id"
        ]
    )

    target_id = (
        selection[
            "target_evidence_id"
        ]
    )


    case_record = (
        golden_case_records_by_id[
            case_id
        ]
    )

    task_plan = next(
        plan
        for plan
        in task_draft_plan_rows
        if (
            plan[
                "task_draft_id"
            ]
            ==
            task_draft_id
        )
    )

    target_record = (
        target_evidence_by_task_draft_id[
            task_draft_id
        ]
    )


    # --------------------------------------------------------
    # Recover actual target text exactly from validated spans.
    # --------------------------------------------------------

    target_text_parts = []


    for part in (
        target_record[
            "target_parts"
        ]
    ):

        document_id = (
            part[
                "document_id"
            ]
        )

        source_text = (
            corpus_target_index[
                document_id
            ][
                "text"
            ]
        )

        char_start = (
            part[
                "char_start"
            ]
        )

        char_end = (
            part[
                "char_end"
            ]
        )

        target_text = source_text[
            char_start:
            char_end
        ]


        # independent span hash check

        recomputed_sha = hashlib.sha256(
            target_text.encode(
                "utf-8"
            )
        ).hexdigest()


        assert (
            recomputed_sha
            ==
            part[
                "target_text_sha256"
            ]
        )


        target_text_parts.append(
            {
                "document_id":
                    document_id,

                "role":
                    part[
                        "role"
                    ],

                "char_start":
                    char_start,

                "char_end":
                    char_end,

                "text":
                    target_text,

                "text_sha256":
                    recomputed_sha,

                "locator":
                    part[
                        "locator"
                    ],

                "attribution":
                    part[
                        "attribution"
                    ],

                "speaker_state":
                    part[
                        "speaker_state"
                    ],
            }
        )


    pilot_case_id = (
        build_pilot_case_id(
            source_type=(
                selection[
                    "source_type"
                ]
            ),
            stratum=(
                selection[
                    "stratum"
                ]
            ),
            golden_case_id=(
                case_id
            ),
            target_evidence_id=(
                target_id
            ),
        )
    )


    packet = {
        "pilot_case_id":
            pilot_case_id,

        "pilot_version":
            QUESTION_DRAFTING_PILOT_VERSION,

        "golden_case_id":
            case_id,

        "task_draft_id":
            task_draft_id,

        "target_evidence_id":
            target_id,

        "source_type":
            selection[
                "source_type"
            ],

        "stratum":
            selection[
                "stratum"
            ],

        "task_family":
            task_plan[
                "task_family"
            ],

        "question_archetype":
            task_plan[
                "question_archetype"
            ],

        "required_dimensions":
            list(
                task_plan[
                    "required_dimensions"
                ]
            ),

        "authority_expectation":
            task_plan[
                "authority_expectation"
            ],

        "context_expectation":
            task_plan[
                "context_expectation"
            ],

        "evidence_scope":
            task_plan[
                "evidence_scope"
            ],

        "target_selection_method":
            task_plan[
                "evidence_target_method"
            ],

        "target_selection_basis":
            target_record[
                "target_selection_basis"
            ],

        "target_total_tokens":
            target_record[
                "target_total_tokens"
            ],

        "target_text_parts":
            target_text_parts,

        "candidate_identity":
            case_record[
                "candidate_identity"
            ],

        "guardrails": {
            "external_knowledge":
                "FORBIDDEN",

            "unknown_speaker_inference":
                "FORBIDDEN",

            "unsupported_attribution":
                "FORBIDDEN",

            "secondary_to_primary_promotion":
                "FORBIDDEN",

            "locator_fabrication":
                "FORBIDDEN",

            "answer_leakage_in_question":
                "FORBIDDEN",

            "internal_ids_in_question":
                "FORBIDDEN",
        },

        "question_draft_status":
            "NOT_STARTED",

        "question_groundability_status":
            "NOT_STARTED",

        "expected_answer_status":
            "NOT_STARTED",

        "citation_target_validation_status":
            "NOT_STARTED",
    }


    question_drafting_pilot_packets.append(
        packet
    )


    if (
        pilot_case_id
        in question_drafting_pilot_by_id
    ):

        raise AssertionError(
            "Duplicate pilot_case_id: "
            f"{pilot_case_id}"
        )


    question_drafting_pilot_by_id[
        pilot_case_id
    ] = (
        packet
    )


# ============================================================
# 6) Independent selection recomputation
# ============================================================

pilot_recomputation_failures = []


for selection in (
    pilot_selection_rows
):

    source_type = (
        selection[
            "source_type"
        ]
    )

    stratum = (
        selection[
            "stratum"
        ]
    )

    targets = (
        targets_by_stratum[
            (
                source_type,
                stratum,
            )
        ]
    )


    token_values = [
        target[
            "target_total_tokens"
        ]
        for target
        in targets
    ]


    median_tokens = (
        empirical_nearest_rank(
            token_values,
            0.50,
        )
    )


    recomputed_candidates = []


    for target in (
        targets
    ):

        tie_hash = hashlib.sha256(
            stable_json_key(
                {
                    "source_type":
                        source_type,

                    "stratum":
                        stratum,

                    "target_evidence_id":
                        target[
                            "target_evidence_id"
                        ],
                }
            ).encode(
                "utf-8"
            )
        ).hexdigest()


        recomputed_candidates.append(
            (
                abs(
                    target[
                        "target_total_tokens"
                    ]
                    -
                    median_tokens
                ),
                tie_hash,
                target[
                    "target_evidence_id"
                ],
            )
        )


    recomputed_winner = sorted(
        recomputed_candidates
    )[0]


    if (
        recomputed_winner[
            2
        ]
        !=
        selection[
            "target_evidence_id"
        ]
    ):

        pilot_recomputation_failures.append(
            {
                "source_type":
                    source_type,

                "stratum":
                    stratum,
            }
        )


# ============================================================
# 7) Pilot integrity audits
# ============================================================

pilot_stratum_counts = Counter(
    (
        packet[
            "source_type"
        ],
        packet[
            "stratum"
        ],
    )
    for packet
    in question_drafting_pilot_packets
)


pilot_source_counts = Counter(
    packet[
        "source_type"
    ]
    for packet
    in question_drafting_pilot_packets
)


pilot_archetype_counts = Counter(
    packet[
        "question_archetype"
    ]
    for packet
    in question_drafting_pilot_packets
)


pilot_empty_target_failures = []

pilot_text_hash_failures = []

pilot_review_leakage = []


for packet in (
    question_drafting_pilot_packets
):

    case_record = (
        golden_case_records_by_id[
            packet[
                "golden_case_id"
            ]
        ]
    )


    if (
        case_record[
            "final_golden_membership"
        ]
        is not True
    ):

        pilot_review_leakage.append(
            packet[
                "pilot_case_id"
            ]
        )


    for part in (
        packet[
            "target_text_parts"
        ]
    ):

        if not (
            part[
                "text"
            ]
            or ""
        ).strip():

            pilot_empty_target_failures.append(
                packet[
                    "pilot_case_id"
                ]
            )


        recomputed_sha = hashlib.sha256(
            part[
                "text"
            ].encode(
                "utf-8"
            )
        ).hexdigest()


        if (
            recomputed_sha
            !=
            part[
                "text_sha256"
            ]
        ):

            pilot_text_hash_failures.append(
                packet[
                    "pilot_case_id"
                ]
            )


# ============================================================
# 8) Underlying target overlap audit
#
# Overlap is descriptive only and is not an automatic failure.
# ============================================================

pilot_span_signatures = Counter()


for packet in (
    question_drafting_pilot_packets
):

    signature = tuple(
        sorted(
            (
                part[
                    "document_id"
                ],
                part[
                    "char_start"
                ],
                part[
                    "char_end"
                ],
            )
            for part
            in packet[
                "target_text_parts"
            ]
        )
    )

    pilot_span_signatures[
        signature
    ] += 1


repeated_pilot_target_spans = {
    signature:
        count

    for signature, count
    in pilot_span_signatures.items()

    if count > 1
}


# ============================================================
# 9) Pilot Registry
# ============================================================

QUESTION_DRAFTING_PILOT_V1_DRAFT = {
    "version":
        QUESTION_DRAFTING_PILOT_VERSION,

    "status":
        "PILOT_COHORT_SELECTED_AND_PACKETS_BUILT",

    "selection_basis":
        "ONE_CASE_PER_READY_STRATUM_CLOSEST_TO_EMPIRICAL_TARGET_TOKEN_MEDIAN",

    "pilot_case_count":
        len(
            question_drafting_pilot_packets
        ),

    "ready_stratum_count":
        len(
            ready_stratum_rows
        ),

    "review_strata_excluded":
        4,

    "source_counts":
        dict(
            pilot_source_counts
        ),

    "question_archetype_counts":
        dict(
            pilot_archetype_counts
        ),

    "repeated_underlying_target_span_count":
        len(
            repeated_pilot_target_spans
        ),

    "question_generation_status":
        "NOT_STARTED",

    "expected_answer_generation_status":
        "NOT_STARTED",

    "artifact_write_status":
        "NONE",
}


# ============================================================
# 10) Validation gates
# ============================================================

exact_23_pilot_cases_gate = (
    len(
        question_drafting_pilot_packets
    )
    == 23
)


exact_one_case_per_ready_stratum_gate = (
    len(
        pilot_stratum_counts
    )
    == 23

    and

    all(
        count == 1
        for count
        in pilot_stratum_counts.values()
    )
)


all_ready_strata_covered_gate = (
    set(
        pilot_stratum_counts.keys()
    )
    ==
    {
        (
            row[
                "source_type"
            ],
            row[
                "stratum"
            ],
        )
        for row
        in ready_stratum_rows
    }
)


pilot_ids_unique_gate = (
    len(
        question_drafting_pilot_by_id
    )
    ==
    len(
        question_drafting_pilot_packets
    )
)


pilot_recomputation_gate = (
    len(
        pilot_recomputation_failures
    )
    == 0
)


no_review_leakage_gate = (
    len(
        pilot_review_leakage
    )
    == 0
)


all_target_text_nonempty_gate = (
    len(
        pilot_empty_target_failures
    )
    == 0
)


all_target_text_hashes_valid_gate = (
    len(
        pilot_text_hash_failures
    )
    == 0
)


all_packets_have_required_task_metadata_gate = all(
    bool(
        packet[
            "task_family"
        ]
    )

    and

    bool(
        packet[
            "question_archetype"
        ]
    )

    and

    bool(
        packet[
            "required_dimensions"
        ]
    )

    and

    bool(
        packet[
            "authority_expectation"
        ]
    )

    and

    bool(
        packet[
            "context_expectation"
        ]
    )

    for packet
    in question_drafting_pilot_packets
)


question_generation_not_started_gate = all(
    packet[
        "question_draft_status"
    ]
    == "NOT_STARTED"

    for packet
    in question_drafting_pilot_packets
)


expected_answer_not_started_gate = all(
    packet[
        "expected_answer_status"
    ]
    == "NOT_STARTED"

    for packet
    in question_drafting_pilot_packets
)


no_artifact_write_gate = (
    QUESTION_DRAFTING_PILOT_V1_DRAFT[
        "artifact_write_status"
    ]
    == "NONE"
)


cell_29_gates = {
    "exact_23_pilot_cases":
        exact_23_pilot_cases_gate,

    "exactly_one_case_per_ready_stratum":
        exact_one_case_per_ready_stratum_gate,

    "all_23_ready_strata_covered":
        all_ready_strata_covered_gate,

    "pilot_case_ids_unique":
        pilot_ids_unique_gate,

    "independent_pilot_selection_recomputation_matches":
        pilot_recomputation_gate,

    "no_review_case_leaks_into_pilot":
        no_review_leakage_gate,

    "all_pilot_target_text_nonempty":
        all_target_text_nonempty_gate,

    "all_pilot_target_text_hashes_valid":
        all_target_text_hashes_valid_gate,

    "all_pilot_packets_have_task_metadata":
        all_packets_have_required_task_metadata_gate,

    "question_generation_not_started":
        question_generation_not_started_gate,

    "expected_answer_generation_not_started":
        expected_answer_not_started_gate,

    "no_artifact_write":
        no_artifact_write_gate,
}


cell_29_pass = all(
    cell_29_gates.values()
)


# ============================================================
# 11) Output
# ============================================================

print(
    "=== QUESTION DRAFTING PILOT COHORT ==="
)

print(
    "Pilot cases:",
    len(
        question_drafting_pilot_packets
    )
)

print(
    "Ready strata represented:",
    len(
        pilot_stratum_counts
    )
)

print(
    "Review strata excluded:",
    4
)


print(
    "\n=== PILOT SOURCE COUNTS ==="
)

for source_type in [
    "website",
    "book",
    "transcript",
]:

    print(
        "-",
        source_type,
        ":",
        pilot_source_counts[
            source_type
        ],
    )


print(
    "\n=== PILOT QUESTION ARCHETYPES ==="
)

for archetype, count in sorted(
    pilot_archetype_counts.items()
):

    print(
        "-",
        archetype,
        ":",
        count,
    )


print(
    "\n=== PER-STRATUM PILOT SELECTION ==="
)

for selection in (
    pilot_selection_rows
):

    print(
        "-",
        selection[
            "source_type"
        ],
        "/",
        selection[
            "stratum"
        ],
        "| cases:",
        selection[
            "stratum_case_count"
        ],
        "| median_tokens:",
        selection[
            "stratum_target_token_median"
        ],
        "| selected_tokens:",
        selection[
            "selected_target_total_tokens"
        ],
        "| distance:",
        selection[
            "distance_to_median"
        ],
        "| case:",
        selection[
            "golden_case_id"
        ],
    )


print(
    "\n=== REPRESENTATIVE DRAFTING PACKETS ==="
)

for packet in (
    question_drafting_pilot_packets
):

    text_previews = [
        {
            "role":
                part[
                    "role"
                ],

            "document_id":
                part[
                    "document_id"
                ],

            "text_preview":
                part[
                    "text"
                ][
                    :500
                ],
        }

        for part
        in packet[
            "target_text_parts"
        ][
            :3
        ]
    ]


    print(
        json.dumps(
            {
                "pilot_case_id":
                    packet[
                        "pilot_case_id"
                    ],

                "source_type":
                    packet[
                        "source_type"
                    ],

                "stratum":
                    packet[
                        "stratum"
                    ],

                "task_family":
                    packet[
                        "task_family"
                    ],

                "question_archetype":
                    packet[
                        "question_archetype"
                    ],

                "required_dimensions":
                    packet[
                        "required_dimensions"
                    ],

                "target_total_tokens":
                    packet[
                        "target_total_tokens"
                    ],

                "target_text_previews":
                    text_previews,

                "question_draft_status":
                    packet[
                        "question_draft_status"
                    ],
            },
            ensure_ascii=False,
            sort_keys=True,
        )
    )


print(
    "\n=== PILOT OVERLAP AUDIT ==="
)

print(
    "Repeated underlying target span signatures:",
    len(
        repeated_pilot_target_spans
    )
)

print(
    "Cross-stratum repetition policy:",
    "DESCRIPTIVE_ONLY_NOT_AUTO_REJECTED"
)


print(
    "\n=== CODE CELL 29 GATES ==="
)

for name, passed in (
    cell_29_gates.items()
):

    print(
        name,
        ":",
        "PASS"
        if passed
        else "FAIL"
    )


print(
    "\nCODE CELL 29 GATE:",
    "PASS"
    if cell_29_pass
    else "FAIL"
)


assert cell_29_pass, (
    "Question-drafting pilot construction failed. "
    "Do not generate pilot questions until all 23 ready strata "
    "are represented by valid, deterministic drafting packets."
)
```

<h3>Full Pilot Evidence Inspection Before Question Drafting</h3>

<p>
Pilot Cohort با موفقیت ساخته شد، اما انتخاب Representative Case بر اساس Median Target Size
فقط نمایندگی اندازه‌ای را تضمین می‌کند؛ این معیار به‌تنهایی ثابت نمی‌کند که Target انتخاب‌شده
برای ساخت یک سؤال طبیعی و مفید مناسب است.
</p>

<p>
در Previewهای Cell 29 چند Case مشاهده شد که ممکن است از نظر Semantic Questionability
نیازمند بررسی بیشتر باشند. برای مثال یک Book Region بسیار کوتاه یا یک Heading منفرد
ممکن است از نظر Retrieval معتبر باشد، اما الزاماً Evidence کافی برای ساخت یک سؤال
Golden طبیعی فراهم نکند.

<p>
بنابراین قبل از تولید اولین Question، متن کامل هر ۲۳ Pilot Target همراه با

<p>
این Cell هیچ Case را Reject یا Replace نمی‌کند و هیچ Threshold معنایی اختراع نمی‌کند.
فقط Evidence واقعی را برای تصمیم مرحله بعد آشکار می‌کند.
</p>

<p>
خروجی این مرحله باید بتواند برای هر Pilot Case پاسخ دهد:
آیا Target از نظر محتوا، ساختار و Attribution واقعاً برای Question Drafting قابل بررسی است،
یا باید قبل از Drafting Target/Pilot آن Case Repair شود؟
</p>

<p>
Question Generation و Expected Answer Generation همچنان شروع نمی‌شوند.
</p>

</div>

```python
# Code Cell 30

# ============================================================
# Full Pilot Evidence Inspection
#
# INSPECTION ONLY.
#
# NO semantic acceptance decision.
# NO pilot replacement.
# NO question generation.
# NO expected-answer generation.
# NO artifact writes.
# ============================================================


# ============================================================
# 1) Build complete inspection records
# ============================================================

pilot_full_inspection_rows = []


for packet in (
    question_drafting_pilot_packets
):

    case_id = (
        packet[
            "golden_case_id"
        ]
    )

    case_record = (
        golden_case_records_by_id[
            case_id
        ]
    )

    candidate = (
        case_record[
            "candidate_snapshot"
        ]
    )

    target_record = (
        target_evidence_by_case_id[
            case_id
        ]
    )


    full_parts = []


    for part_index, part in enumerate(
        packet[
            "target_text_parts"
        ],
        start=1,
    ):

        document_id = (
            part[
                "document_id"
            ]
        )

        corpus_record = (
            corpus_target_index[
                document_id
            ]
        )

        source_metadata = (
            corpus_record[
                "source_metadata"
            ]
            or {}
        )

        locator = (
            corpus_record[
                "locator"
            ]
            or {}
        )

        text = (
            part[
                "text"
            ]
        )


        full_parts.append(
            {
                "part_index":
                    part_index,

                "role":
                    part[
                        "role"
                    ],

                "document_id":
                    document_id,

                "source_id":
                    corpus_record[
                        "source_id"
                    ],

                "title":
                    corpus_record[
                        "title"
                    ],

                "char_start":
                    part[
                        "char_start"
                    ],

                "char_end":
                    part[
                        "char_end"
                    ],

                "char_count":
                    len(
                        text
                    ),

                "token_count":
                    nb11_token_count(
                        text
                    ),

                "question_mark_count":
                    (
                        text.count(
                            "?"
                        )
                        +
                        text.count(
                            "؟"
                        )
                    ),

                "newline_count":
                    text.count(
                        "\n"
                    ),

                "locator":
                    locator,

                "source_metadata_subset": {
                    key:
                        source_metadata.get(
                            key
                        )

                    for key in [
                        "paragraph_id",
                        "paragraph_number",
                        "heading_path",
                        "heading_depth",
                        "is_heading",
                        "in_table",
                        "book_unit_type",
                        "content_type",
                        "program_id",
                        "episode_number",
                        "season_number",
                        "utterance_index",
                    ]

                    if (
                        key
                        in source_metadata
                    )
                },

                "attribution":
                    part[
                        "attribution"
                    ],

                "speaker_state":
                    part[
                        "speaker_state"
                    ],

                "text":
                    text,
            }
        )


    joined_target_text = "\n\n".join(
        part[
            "text"
        ]
        for part
        in full_parts
    )


    inspection_row = {
        "pilot_case_id":
            packet[
                "pilot_case_id"
            ],

        "golden_case_id":
            case_id,

        "source_type":
            packet[
                "source_type"
            ],

        "stratum":
            packet[
                "stratum"
            ],

        "task_family":
            packet[
                "task_family"
            ],

        "question_archetype":
            packet[
                "question_archetype"
            ],

        "required_dimensions":
            packet[
                "required_dimensions"
            ],

        "authority_expectation":
            packet[
                "authority_expectation"
            ],

        "context_expectation":
            packet[
                "context_expectation"
            ],

        "target_selection_method":
            packet[
                "target_selection_method"
            ],

        "target_selection_basis":
            packet[
                "target_selection_basis"
            ],

        "candidate_identity":
            packet[
                "candidate_identity"
            ],

        "candidate_snapshot":
            candidate,

        "target_part_count":
            len(
                full_parts
            ),

        "target_total_tokens":
            target_record[
                "target_total_tokens"
            ],

        "joined_target_char_count":
            len(
                joined_target_text
            ),

        "joined_question_mark_count":
            (
                joined_target_text.count(
                    "?"
                )
                +
                joined_target_text.count(
                    "؟"
                )
            ),

        "target_parts":
            full_parts,

        # Deliberately unresolved.
        "semantic_questionability_status":
            "NOT_YET_DECIDED",

        "question_draft_status":
            "NOT_STARTED",

        "expected_answer_status":
            "NOT_STARTED",
    }


    pilot_full_inspection_rows.append(
        inspection_row
    )


# ============================================================
# 2) Structural distributions
#
# Descriptive only.
# No semantic thresholds are created.
# ============================================================

pilot_token_distribution = sorted(
    row[
        "target_total_tokens"
    ]
    for row
    in pilot_full_inspection_rows
)


pilot_part_distribution = Counter(
    row[
        "target_part_count"
    ]
    for row
    in pilot_full_inspection_rows
)


pilot_question_mark_distribution = Counter(
    row[
        "joined_question_mark_count"
    ]
    for row
    in pilot_full_inspection_rows
)


pilot_rows_by_token_size = sorted(
    pilot_full_inspection_rows,
    key=lambda row: (
        row[
            "target_total_tokens"
        ],
        row[
            "source_type"
        ],
        row[
            "stratum"
        ],
    ),
)


# ============================================================
# 3) Independent target-text integrity validation
# ============================================================

pilot_full_text_integrity_failures = []

pilot_target_id_linkage_failures = []

pilot_case_linkage_failures = []


for row in (
    pilot_full_inspection_rows
):

    case_id = (
        row[
            "golden_case_id"
        ]
    )

    packet = next(
        packet
        for packet
        in question_drafting_pilot_packets
        if (
            packet[
                "pilot_case_id"
            ]
            ==
            row[
                "pilot_case_id"
            ]
        )
    )

    target_record = (
        target_evidence_by_case_id[
            case_id
        ]
    )


    if (
        packet[
            "target_evidence_id"
        ]
        !=
        target_record[
            "target_evidence_id"
        ]
    ):

        pilot_target_id_linkage_failures.append(
            row[
                "pilot_case_id"
            ]
        )


    if (
        packet[
            "golden_case_id"
        ]
        !=
        row[
            "golden_case_id"
        ]
    ):

        pilot_case_linkage_failures.append(
            row[
                "pilot_case_id"
            ]
        )


    for part in (
        row[
            "target_parts"
        ]
    ):

        document_id = (
            part[
                "document_id"
            ]
        )

        source_text = (
            corpus_target_index[
                document_id
            ][
                "text"
            ]
        )

        recovered = source_text[
            part[
                "char_start"
            ]:
            part[
                "char_end"
            ]
        ]

        recovered_sha = hashlib.sha256(
            recovered.encode(
                "utf-8"
            )
        ).hexdigest()


        if (
            recovered
            !=
            part[
                "text"
            ]

            or

            recovered_sha
            !=
            hashlib.sha256(
                part[
                    "text"
                ].encode(
                    "utf-8"
                )
            ).hexdigest()
        ):

            pilot_full_text_integrity_failures.append(
                {
                    "pilot_case_id":
                        row[
                            "pilot_case_id"
                        ],

                    "document_id":
                        document_id,
                }
            )


# ============================================================
# 4) Explicitly preserve semantic undecided state
# ============================================================

semantic_decision_counts = Counter(
    row[
        "semantic_questionability_status"
    ]
    for row
    in pilot_full_inspection_rows
)


# ============================================================
# 5) Inspection Registry
# ============================================================

PILOT_FULL_EVIDENCE_INSPECTION_V1_DRAFT = {
    "version":
        "pilot_full_evidence_inspection_v1_draft",

    "status":
        "FULL_EVIDENCE_EXPOSED_FOR_SEMANTIC_REVIEW",

    "pilot_case_count":
        len(
            pilot_full_inspection_rows
        ),

    "semantic_decision_policy":
        (
            "NO_AUTOMATIC_SEMANTIC_ACCEPTANCE_FROM_TARGET_SIZE, "
            "TOKEN_COUNT, OR STRUCTURAL VALIDITY"
        ),

    "semantic_questionability_status":
        "NOT_YET_DECIDED",

    "question_generation_status":
        "NOT_STARTED",

    "expected_answer_generation_status":
        "NOT_STARTED",

    "artifact_write_status":
        "NONE",
}


# ============================================================
# 6) Validation gates
# ============================================================

exact_23_full_inspection_rows_gate = (
    len(
        pilot_full_inspection_rows
    )
    == 23
)


all_pilot_ids_unique_gate = (
    len(
        {
            row[
                "pilot_case_id"
            ]
            for row
            in pilot_full_inspection_rows
        }
    )
    == 23
)


all_targets_nonempty_gate = all(
    row[
        "target_part_count"
    ]
    > 0

    and

    row[
        "joined_target_char_count"
    ]
    > 0

    and

    row[
        "target_total_tokens"
    ]
    > 0

    for row
    in pilot_full_inspection_rows
)


target_text_integrity_gate = (
    len(
        pilot_full_text_integrity_failures
    )
    == 0
)


target_id_linkage_gate = (
    len(
        pilot_target_id_linkage_failures
    )
    == 0
)


case_linkage_gate = (
    len(
        pilot_case_linkage_failures
    )
    == 0
)


semantic_decision_not_prejudged_gate = (
    semantic_decision_counts
    ==
    Counter(
        {
            "NOT_YET_DECIDED":
                23
        }
    )
)


question_generation_not_started_gate = all(
    row[
        "question_draft_status"
    ]
    == "NOT_STARTED"

    for row
    in pilot_full_inspection_rows
)


expected_answer_not_started_gate = all(
    row[
        "expected_answer_status"
    ]
    == "NOT_STARTED"

    for row
    in pilot_full_inspection_rows
)


no_artifact_write_gate = (
    PILOT_FULL_EVIDENCE_INSPECTION_V1_DRAFT[
        "artifact_write_status"
    ]
    == "NONE"
)


cell_30_gates = {
    "exact_23_pilot_cases_fully_inspected":
        exact_23_full_inspection_rows_gate,

    "all_pilot_case_ids_unique":
        all_pilot_ids_unique_gate,

    "all_full_targets_nonempty":
        all_targets_nonempty_gate,

    "full_target_text_integrity":
        target_text_integrity_gate,

    "target_evidence_id_linkage_integrity":
        target_id_linkage_gate,

    "golden_case_linkage_integrity":
        case_linkage_gate,

    "semantic_questionability_not_prejudged":
        semantic_decision_not_prejudged_gate,

    "question_generation_not_started":
        question_generation_not_started_gate,

    "expected_answer_generation_not_started":
        expected_answer_not_started_gate,

    "no_artifact_write":
        no_artifact_write_gate,
}


cell_30_pass = all(
    cell_30_gates.values()
)


# ============================================================
# 7) Output
# ============================================================

print(
    "=== FULL PILOT EVIDENCE INSPECTION ==="
)

print(
    "Pilot cases fully exposed:",
    len(
        pilot_full_inspection_rows
    )
)

print(
    "Semantic questionability decisions:",
    dict(
        semantic_decision_counts
    )
)


print(
    "\n=== PILOT TARGET SIZE DISTRIBUTION ==="
)

print(
    "min:",
    min(
        pilot_token_distribution
    )
)

print(
    "median:",
    empirical_nearest_rank(
        pilot_token_distribution,
        0.50,
    )
)

print(
    "max:",
    max(
        pilot_token_distribution
    )
)


print(
    "\n=== TARGET PART COUNT DISTRIBUTION ==="
)

for part_count, count in sorted(
    pilot_part_distribution.items()
):

    print(
        "-",
        part_count,
        "parts:",
        count,
        "cases",
    )


print(
    "\n=== SMALLEST PILOT TARGETS ==="
)

for row in (
    pilot_rows_by_token_size[
        :8
    ]
):

    print(
        "-",
        row[
            "source_type"
        ],
        "/",
        row[
            "stratum"
        ],
        "| tokens:",
        row[
            "target_total_tokens"
        ],
        "| parts:",
        row[
            "target_part_count"
        ],
        "| case:",
        row[
            "golden_case_id"
        ],
    )


print(
    "\n=== COMPLETE PILOT REVIEW PACKETS ==="
)


for index, row in enumerate(
    pilot_full_inspection_rows,
    start=1,
):

    print(
        "\n"
        + "=" * 90
    )

    print(
        "PILOT",
        index,
        "/",
        len(
            pilot_full_inspection_rows
        )
    )

    print(
        "Source / Stratum:",
        row[
            "source_type"
        ],
        "/",
        row[
            "stratum"
        ]
    )

    print(
        "Pilot Case ID:",
        row[
            "pilot_case_id"
        ]
    )

    print(
        "Golden Case ID:",
        row[
            "golden_case_id"
        ]
    )

    print(
        "Task Family:",
        row[
            "task_family"
        ]
    )

    print(
        "Question Archetype:",
        row[
            "question_archetype"
        ]
    )

    print(
        "Required Dimensions:",
        row[
            "required_dimensions"
        ]
    )

    print(
        "Authority Expectation:",
        row[
            "authority_expectation"
        ]
    )

    print(
        "Context Expectation:",
        row[
            "context_expectation"
        ]
    )

    print(
        "Target Selection Method:",
        row[
            "target_selection_method"
        ]
    )

    print(
        "Target Selection Basis:",
        row[
            "target_selection_basis"
        ]
    )

    print(
        "Target Parts:",
        row[
            "target_part_count"
        ]
    )

    print(
        "Target Tokens:",
        row[
            "target_total_tokens"
        ]
    )

    print(
        "Candidate Snapshot:"
    )

    print(
        json.dumps(
            row[
                "candidate_snapshot"
            ],
            ensure_ascii=False,
            sort_keys=True,
        )
    )


    for part in (
        row[
            "target_parts"
        ]
    ):

        print(
            "\n--- TARGET PART",
            part[
                "part_index"
            ],
            "---"
        )

        print(
            "Role:",
            part[
                "role"
            ]
        )

        print(
            "Document ID:",
            part[
                "document_id"
            ]
        )

        print(
            "Title:",
            part[
                "title"
            ]
        )

        print(
            "Token Count:",
            part[
                "token_count"
            ]
        )

        print(
            "Question Marks:",
            part[
                "question_mark_count"
            ]
        )

        print(
            "Source Metadata:"
        )

        print(
            json.dumps(
                part[
                    "source_metadata_subset"
                ],
                ensure_ascii=False,
                sort_keys=True,
            )
        )

        print(
            "Locator:"
        )

        print(
            json.dumps(
                part[
                    "locator"
                ],
                ensure_ascii=False,
                sort_keys=True,
            )
        )

        print(
            "Attribution:"
        )

        print(
            json.dumps(
                part[
                    "attribution"
                ],
                ensure_ascii=False,
                sort_keys=True,
            )
        )

        print(
            "Speaker State:"
        )

        print(
            json.dumps(
                part[
                    "speaker_state"
                ],
                ensure_ascii=False,
                sort_keys=True,
            )
        )

        print(
            "FULL TARGET TEXT:"
        )

        print(
            part[
                "text"
            ]
        )


    print(
        "\nSEMANTIC QUESTIONABILITY:",
        row[
            "semantic_questionability_status"
        ]
    )


print(
    "\n=== INTEGRITY FAILURES ==="
)

print(
    "Full text integrity failures:",
    len(
        pilot_full_text_integrity_failures
    )
)

print(
    "Target ID linkage failures:",
    len(
        pilot_target_id_linkage_failures
    )
)

print(
    "Case linkage failures:",
    len(
        pilot_case_linkage_failures
    )
)


print(
    "\n=== CODE CELL 30 GATES ==="
)

for name, passed in (
    cell_30_gates.items()
):

    print(
        name,
        ":",
        "PASS"
        if passed
        else "FAIL"
    )


print(
    "\nCODE CELL 30 GATE:",
    "PASS"
    if cell_30_pass
    else "FAIL"
)


assert cell_30_pass, (
    "Full Pilot Evidence Inspection failed. "
    "Do not draft questions until the complete target evidence "
    "for all 23 pilot cases is visible and integrity-validated."
)
```

<h3>Semantic Pilot Adjudication and Repair Gate</h3>

<p>
Structural Validity و Evidence Integrity ثابت نمی‌کنند که یک Target
برای ساخت سؤال Golden مناسب است.
Cell 30 متن کامل ۲۳ Pilot را آشکار کرد و اکنون می‌توان Semantic Readiness
را مستقل از صحت ساختاری بررسی کرد.
</p>

<p>
این Adjudication یک Human Confirmation برای چهار Review Stratum نیست.
آن چهار Stratum همچنان جداگانه Block هستند.
این مرحله فقط کیفیت Semantic Targetهای Pilot Auto-ready را بررسی می‌کند.
</p>

<h4>معیارهای Semantic Readiness</h4>

<ul>
<li>Target باید برای Task Family مربوطه محتوای کافی داشته باشد.</li>
<li>سؤال آینده باید بدون External Knowledge از Target قابل ساخت باشد.</li>
<li>Target نباید صرفاً یک Label، App Name یا Fragment فاقد محتوای کافی باشد، مگر Task عمداً Locator-focused باشد.</li>
<li>برای RAG Finance، Content-retrieval Targets باید با موضوعات مالی، سواد مالی، تصمیم‌گیری مالی یا زمینه آموزشی مرتبط باشند؛ صرف حضور در Corpus برای انتخاب Semantic Target کافی نیست.</li>
<li>Attribution/Authority Boundary Targets می‌توانند شامل دو Speaker State متفاوت باشند و لازم نیست هر دو بخش محتوای مستقل و هم‌اندازه داشته باشند.</li>
<li>Unresolved-speaker Case عمداً می‌تواند Question Abstention را ارزیابی کند و نباید Speaker Identity برای آن ساخته شود.</li>
<li>Deep Heading Case می‌تواند متن کوتاه داشته باشد، چون هدف اصلی آن Hierarchical Locator Retrieval است.</li>
</ul>

<p>
نتیجه‌ی این Cell فقط سه حالت خواهد داشت:
</p>

<ul>
<li><code>READY_FOR_PILOT_QUESTION_DRAFT</code></li>
<li><code>READY_SPECIAL_ABSTENTION_OR_LOCATOR_TASK</code></li>
<li><code>REPAIR_REQUIRED_BEFORE_QUESTION_DRAFT</code></li>
</ul>

<p>
هیچ Question یا Expected Answer در این مرحله تولید نمی‌شود.
</p>

```python
# Code Cell 31

# ============================================================
# Semantic Pilot Adjudication
#
# Explicit authoring-review decisions derived from the complete
# evidence exposed in Code Cell 30.
#
# This is NOT human confirmation for the 4 review-required
# strata from the Golden Dataset contract.
#
# NO question generation.
# NO expected-answer generation.
# NO artifact writes.
# ============================================================


# ============================================================
# 1) Semantic review contract
# ============================================================

PILOT_SEMANTIC_REVIEW_CONTRACT_V1_DRAFT = {
    "version":
        "pilot_semantic_review_contract_v1_draft",

    "status":
        "AUTHORING_REVIEW_IN_MEMORY_NOT_FROZEN",

    "criteria": {
        "task_relevance_required":
            True,

        "evidence_sufficiency_required":
            True,

        "external_knowledge_required":
            False,

        "content_retrieval_finance_or_financial_education_relevance_required":
            True,

        "short_target_automatically_invalid":
            False,

        "locator_task_may_use_short_target":
            True,

        "abstention_task_may_preserve_unresolved_identity":
            True,

        "attribution_boundary_may_use_asymmetric_pair":
            True,
    },

    "allowed_statuses": [
        "READY_FOR_PILOT_QUESTION_DRAFT",
        "READY_SPECIAL_ABSTENTION_OR_LOCATOR_TASK",
        "REPAIR_REQUIRED_BEFORE_QUESTION_DRAFT",
    ],

    "question_generation_status":
        "NOT_STARTED",

    "expected_answer_generation_status":
        "NOT_STARTED",

    "artifact_write_status":
        "NONE",
}


# ============================================================
# 2) Explicit semantic adjudication
#
# Important:
# - Decisions are based on full evidence from Cell 30.
# - No token threshold is used.
# - No automatic semantic classifier is introduced.
# ============================================================

PILOT_SEMANTIC_ADJUDICATION = {

    # --------------------------------------------------------
    # WEBSITE
    # --------------------------------------------------------

    ("website", "direct_author_short"): {
        "status":
            "READY_FOR_PILOT_QUESTION_DRAFT",

        "reason":
            (
                "Substantive financial-market content is directly "
                "grounded and attribution is explicit."
            ),
    },


    ("website", "direct_author_long"): {
        "status":
            "READY_FOR_PILOT_QUESTION_DRAFT",

        "reason":
            (
                "Local target contains explicit financial-literacy "
                "concepts and supports an evidence-completeness task."
            ),
    },


    ("website", "supervised_short"): {
        "status":
            "READY_FOR_PILOT_QUESTION_DRAFT",

        "reason":
            (
                "Substantive financial-literacy research content is "
                "available and supervised-secondary authority is explicit."
            ),
    },


    ("website", "supervised_long"): {
        "status":
            "READY_FOR_PILOT_QUESTION_DRAFT",

        "reason":
            (
                "Target contains coherent saving-related educational "
                "content and remains distinguishable from direct authorship."
            ),
    },


    # --------------------------------------------------------
    # BOOK
    # --------------------------------------------------------

    ("book", "intrinsically_tiny_narrative_region"): {
        "status":
            "READY_FOR_PILOT_QUESTION_DRAFT",

        "reason":
            (
                "The protected region contains a complete exercise "
                "instruction despite its small size."
            ),
    },


    ("book", "intrinsically_tiny_table_region"): {
        "status":
            "READY_FOR_PILOT_QUESTION_DRAFT",

        "reason":
            (
                "The complete protected table region contains interpretable "
                "resource/title entries and preserves table structure."
            ),
    },


    ("book", "long_section"): {
        "status":
            "READY_FOR_PILOT_QUESTION_DRAFT",

        "reason":
            (
                "Two adjacent paragraphs jointly define financial-literacy "
                "standards and their knowledge/practical structure."
            ),
    },


    ("book", "large_table_run"): {
        "status":
            "READY_FOR_PILOT_QUESTION_DRAFT",

        "reason":
            (
                "The table run is semantically coherent as a job-satisfaction "
                "worksheet and supports structure-aware retrieval."
            ),
    },


    ("book", "exercise_or_self_assessment"): {
        "status":
            "REPAIR_REQUIRED_BEFORE_QUESTION_DRAFT",

        "reason":
            (
                "The selected target is only the text "
                "'برنامه کاربردی «سیگنال»'. "
                "The worksheet heading path exists, but the selected record "
                "does not contain enough exercise/self-assessment content "
                "for a grounded evaluation question."
            ),

        "repair_scope":
            "TARGET_SELECTION_METHOD_OR_PILOT_CASE",

        "suspected_systemic_method":
            "USE_EXPLICIT_BOOK_REGION",
    },


    ("book", "supplemental_footnote_case"): {
        "status":
            "READY_FOR_PILOT_QUESTION_DRAFT",

        "reason":
            (
                "Parent paragraph plus supplemental 'product' note forms "
                "a valid parent-footnote evidence relationship."
            ),
    },


    ("book", "deep_heading_path"): {
        "status":
            "READY_SPECIAL_ABSTENTION_OR_LOCATOR_TASK",

        "reason":
            (
                "The target text is short, but this stratum explicitly "
                "tests hierarchical locator retrieval. The question text "
                "'چه گزینه‌هایی پیش‌رو دارم؟' plus its heading path is "
                "sufficient for a locator-focused task."
            ),
    },


    # --------------------------------------------------------
    # TRANSCRIPT
    # --------------------------------------------------------

    ("transcript", "single_speaker_primary"): {
        "status":
            "REPAIR_REQUIRED_BEFORE_QUESTION_DRAFT",

        "reason":
            (
                "Attribution is valid, but the selected longest utterance "
                "is a missile-shelter/sandbag example and does not provide "
                "a suitable financial or financial-literacy target for the "
                "RAG Finance evaluation pilot."
            ),

        "repair_scope":
            "PILOT_TARGET_SELECTION",

        "suspected_systemic_method":
            "DETERMINISTIC_LOCAL_UTTERANCE_TARGET_REQUIRED",
    },


    ("transcript", "single_speaker_unresolved"): {
        "status":
            "READY_SPECIAL_ABSTENTION_OR_LOCATOR_TASK",

        "reason":
            (
                "Financial-literacy content is substantive, while speaker "
                "identity remains explicitly unresolved. This is suitable "
                "for an attribution-abstention evaluation."
            ),
    },


    ("transcript", "multi_speaker_uni_tehran"): {
        "status":
            "READY_FOR_PILOT_QUESTION_DRAFT",

        "reason":
            (
                "Both adjacent utterances contain substantive financial "
                "content with distinct explicit attribution states."
            ),
    },


    ("transcript", "multi_speaker_chehelston"): {
        "status":
            "READY_FOR_PILOT_QUESTION_DRAFT",

        "reason":
            (
                "Primary Dr. Roudi evidence is adjacent to unresolved "
                "speaker evidence, making non-inference and attribution "
                "boundary behavior directly testable."
            ),
    },


    ("transcript", "multi_speaker_asre_shirin"): {
        "status":
            "READY_FOR_PILOT_QUESTION_DRAFT",

        "reason":
            (
                "Dr. Roudi-labelled and participant-labelled utterances "
                "are both substantive and support attribution separation."
            ),
    },


    ("transcript", "multi_speaker_khane_to"): {
        "status":
            "READY_FOR_PILOT_QUESTION_DRAFT",

        "reason":
            (
                "The evidence contains substantive Dr. Roudi financial "
                "content followed by an explicitly labelled host utterance."
            ),
    },


    ("transcript", "multi_speaker_mizan"): {
        "status":
            "READY_SPECIAL_ABSTENTION_OR_LOCATOR_TASK",

        "reason":
            (
                "One utterance is narratively labelled 'یکی از حاضران' "
                "while the response is unresolved. The case is valid only "
                "when the future task preserves that unresolved identity "
                "instead of inventing the responder."
            ),
    },


    ("transcript", "primary_to_nonprimary_transition"): {
        "status":
            "READY_FOR_PILOT_QUESTION_DRAFT",

        "reason":
            (
                "The primary-side utterance is short, but the verified "
                "adjacent transition itself is the evaluation target; "
                "the substantive right-side budget explanation must not "
                "be attributed to Dr. Roudi."
            ),
    },


    ("transcript", "mixed_primary_signal_case"): {
        "status":
            "READY_FOR_PILOT_QUESTION_DRAFT",

        "reason":
            (
                "The pair contains substantial cash-flow discussion with "
                "different primary-authority states and is suitable for "
                "authority-boundary evaluation."
            ),
    },


    ("transcript", "long_utterance_over_500"): {
        "status":
            "READY_FOR_PILOT_QUESTION_DRAFT",

        "reason":
            (
                "The long utterance contains coherent financial reasoning "
                "about borrowing, debt dependence, and financial capability."
            ),
    },


    ("transcript", "long_utterance_over_800"): {
        "status":
            "READY_FOR_PILOT_QUESTION_DRAFT",

        "reason":
            (
                "The utterance provides extensive coherent evidence about "
                "NPV, IRR, discount rates, and investment evaluation."
            ),
    },


    ("transcript", "token_fallback_sentence_case"): {
        "status":
            "READY_FOR_PILOT_QUESTION_DRAFT",

        "reason":
            (
                "The oversized source sentence contains coherent financial "
                "content spanning investment risk management and insurance "
                "terminology, making fragment reconstruction testable."
            ),
    },
}


# ============================================================
# 3) Verify adjudication covers exactly the 23 pilot strata
# ============================================================

pilot_stratum_keys = {
    (
        row[
            "source_type"
        ],
        row[
            "stratum"
        ],
    )
    for row
    in pilot_full_inspection_rows
}


adjudication_keys = set(
    PILOT_SEMANTIC_ADJUDICATION.keys()
)


semantic_coverage_missing = sorted(
    pilot_stratum_keys
    -
    adjudication_keys
)


semantic_coverage_extra = sorted(
    adjudication_keys
    -
    pilot_stratum_keys
)


# ============================================================
# 4) Attach semantic decisions to pilot records
# ============================================================

pilot_semantic_review_rows = []


for row in (
    pilot_full_inspection_rows
):

    key = (
        row[
            "source_type"
        ],
        row[
            "stratum"
        ],
    )

    decision = (
        PILOT_SEMANTIC_ADJUDICATION[
            key
        ]
    )


    semantic_row = {
        "pilot_case_id":
            row[
                "pilot_case_id"
            ],

        "golden_case_id":
            row[
                "golden_case_id"
            ],

        "source_type":
            row[
                "source_type"
            ],

        "stratum":
            row[
                "stratum"
            ],

        "task_family":
            row[
                "task_family"
            ],

        "question_archetype":
            row[
                "question_archetype"
            ],

        "target_total_tokens":
            row[
                "target_total_tokens"
            ],

        "semantic_status":
            decision[
                "status"
            ],

        "semantic_reason":
            decision[
                "reason"
            ],

        "repair_scope":
            decision.get(
                "repair_scope"
            ),

        "suspected_systemic_method":
            decision.get(
                "suspected_systemic_method"
            ),

        "question_draft_status":
            (
                "BLOCKED_FOR_REPAIR"
                if (
                    decision[
                        "status"
                    ]
                    ==
                    "REPAIR_REQUIRED_BEFORE_QUESTION_DRAFT"
                )
                else
                "READY_NOT_STARTED"
            ),

        "expected_answer_status":
            "NOT_STARTED",
    }


    pilot_semantic_review_rows.append(
        semantic_row
    )


# ============================================================
# 5) Summary
# ============================================================

semantic_status_counts = Counter(
    row[
        "semantic_status"
    ]
    for row
    in pilot_semantic_review_rows
)


repair_rows = [
    row
    for row
    in pilot_semantic_review_rows
    if (
        row[
            "semantic_status"
        ]
        ==
        "REPAIR_REQUIRED_BEFORE_QUESTION_DRAFT"
    )
]


special_rows = [
    row
    for row
    in pilot_semantic_review_rows
    if (
        row[
            "semantic_status"
        ]
        ==
        "READY_SPECIAL_ABSTENTION_OR_LOCATOR_TASK"
    )
]


normal_ready_rows = [
    row
    for row
    in pilot_semantic_review_rows
    if (
        row[
            "semantic_status"
        ]
        ==
        "READY_FOR_PILOT_QUESTION_DRAFT"
    )
]


pilot_ready_after_semantic_review = (
    normal_ready_rows
    +
    special_rows
)


suspected_systemic_methods = sorted(
    {
        row[
            "suspected_systemic_method"
        ]
        for row
        in repair_rows
        if (
            row[
                "suspected_systemic_method"
            ]
            is not None
        )
    }
)


# ============================================================
# 6) Semantic Review Registry
# ============================================================

PILOT_SEMANTIC_REVIEW_V1_DRAFT = {
    "version":
        "pilot_semantic_review_v1_draft",

    "status":
        "ADJUDICATED_REPAIRS_REQUIRED",

    "pilot_case_count":
        len(
            pilot_semantic_review_rows
        ),

    "ready_normal_count":
        len(
            normal_ready_rows
        ),

    "ready_special_count":
        len(
            special_rows
        ),

    "repair_required_count":
        len(
            repair_rows
        ),

    "ready_for_question_draft_count":
        len(
            pilot_ready_after_semantic_review
        ),

    "suspected_systemic_target_methods":
        suspected_systemic_methods,

    "question_generation_status":
        "NOT_STARTED",

    "expected_answer_generation_status":
        "NOT_STARTED",

    "artifact_write_status":
        "NONE",
}


# ============================================================
# 7) Validation gates
# ============================================================

exact_23_semantic_decisions_gate = (
    len(
        pilot_semantic_review_rows
    )
    == 23
)


exact_semantic_key_coverage_gate = (
    len(
        semantic_coverage_missing
    )
    == 0

    and

    len(
        semantic_coverage_extra
    )
    == 0
)


all_statuses_valid_gate = all(
    row[
        "semantic_status"
    ]
    in
    PILOT_SEMANTIC_REVIEW_CONTRACT_V1_DRAFT[
        "allowed_statuses"
    ]

    for row
    in pilot_semantic_review_rows
)


repair_cases_blocked_gate = all(
    row[
        "question_draft_status"
    ]
    ==
    "BLOCKED_FOR_REPAIR"

    for row
    in repair_rows
)


ready_cases_not_drafted_gate = all(
    row[
        "question_draft_status"
    ]
    ==
    "READY_NOT_STARTED"

    for row
    in pilot_ready_after_semantic_review
)


repair_reasons_nonempty_gate = all(
    bool(
        row[
            "semantic_reason"
        ]
    )

    and

    bool(
        row[
            "repair_scope"
        ]
    )

    for row
    in repair_rows
)


special_cases_are_explicit_gate = (
    {
        (
            row[
                "source_type"
            ],
            row[
                "stratum"
            ],
        )
        for row
        in special_rows
    }
    ==
    {
        (
            "book",
            "deep_heading_path",
        ),
        (
            "transcript",
            "single_speaker_unresolved",
        ),
        (
            "transcript",
            "multi_speaker_mizan",
        ),
    }
)


repairs_detected_gate = (
    {
        (
            row[
                "source_type"
            ],
            row[
                "stratum"
            ],
        )
        for row
        in repair_rows
    }
    ==
    {
        (
            "book",
            "exercise_or_self_assessment",
        ),
        (
            "transcript",
            "single_speaker_primary",
        ),
    }
)


question_generation_not_started_gate = (
    PILOT_SEMANTIC_REVIEW_V1_DRAFT[
        "question_generation_status"
    ]
    ==
    "NOT_STARTED"

    and

    all(
        row[
            "question_draft_status"
        ]
        in {
            "READY_NOT_STARTED",
            "BLOCKED_FOR_REPAIR",
        }

        for row
        in pilot_semantic_review_rows
    )
)


expected_answer_not_started_gate = all(
    row[
        "expected_answer_status"
    ]
    ==
    "NOT_STARTED"

    for row
    in pilot_semantic_review_rows
)


no_artifact_write_gate = (
    PILOT_SEMANTIC_REVIEW_V1_DRAFT[
        "artifact_write_status"
    ]
    == "NONE"
)


cell_31_gates = {
    "exact_23_pilot_semantic_decisions":
        exact_23_semantic_decisions_gate,

    "semantic_adjudication_covers_exact_pilot_strata":
        exact_semantic_key_coverage_gate,

    "all_semantic_statuses_valid":
        all_statuses_valid_gate,

    "repair_cases_blocked_before_question_draft":
        repair_cases_blocked_gate,

    "semantically_ready_cases_remain_not_drafted":
        ready_cases_not_drafted_gate,

    "all_repair_cases_have_explicit_reason_and_scope":
        repair_reasons_nonempty_gate,

    "three_special_abstention_or_locator_cases_explicit":
        special_cases_are_explicit_gate,

    "two_semantic_repairs_detected":
        repairs_detected_gate,

    "question_generation_not_started":
        question_generation_not_started_gate,

    "expected_answer_generation_not_started":
        expected_answer_not_started_gate,

    "no_artifact_write":
        no_artifact_write_gate,
}


cell_31_pass = all(
    cell_31_gates.values()
)


# ============================================================
# 8) Output
# ============================================================

print(
    "=== PILOT SEMANTIC ADJUDICATION ==="
)

print(
    "Pilot cases reviewed:",
    len(
        pilot_semantic_review_rows
    )
)


print(
    "\n=== SEMANTIC STATUS COUNTS ==="
)

for status, count in sorted(
    semantic_status_counts.items()
):

    print(
        "-",
        status,
        ":",
        count,
    )


print(
    "\nReady for future question drafting:",
    len(
        pilot_ready_after_semantic_review
    )
)

print(
    "Repair required before drafting:",
    len(
        repair_rows
    )
)


print(
    "\n=== REPAIR-REQUIRED PILOTS ==="
)

for row in (
    repair_rows
):

    print(
        "-",
        row[
            "source_type"
        ],
        "/",
        row[
            "stratum"
        ]
    )

    print(
        "   case:",
        row[
            "golden_case_id"
        ]
    )

    print(
        "   scope:",
        row[
            "repair_scope"
        ]
    )

    print(
        "   suspected method:",
        row[
            "suspected_systemic_method"
        ]
    )

    print(
        "   reason:",
        row[
            "semantic_reason"
        ]
    )


print(
    "\n=== SPECIAL READY PILOTS ==="
)

for row in (
    special_rows
):

    print(
        "-",
        row[
            "source_type"
        ],
        "/",
        row[
            "stratum"
        ],
        "| reason:",
        row[
            "semantic_reason"
        ],
    )


print(
    "\n=== SUSPECTED SYSTEMIC TARGET METHODS ==="
)

for method in (
    suspected_systemic_methods
):

    print(
        "-",
        method
    )


print(
    "\n=== CODE CELL 31 GATES ==="
)

for name, passed in (
    cell_31_gates.items()
):

    print(
        name,
        ":",
        "PASS"
        if passed
        else "FAIL"
    )


print(
    "\nCODE CELL 31 GATE:",
    "PASS"
    if cell_31_pass
    else "FAIL"
)


assert cell_31_pass, (
    "Pilot semantic adjudication failed. "
    "Do not draft questions until repair-required pilot targets "
    "are isolated and semantic statuses are internally consistent."
)
```

```python
# Code Cell 32

# ============================================================
# Systemic Repair Scope Audit — REPAIRED
#
# READ / INSPECT ONLY.
#
# Repair note:
# Cell 28 select_transcript_local_utterance() used
# stable_target_tie_hash({"document_id": document_id})
# as the deterministic tie-break.
#
# The previous Cell 32 incorrectly used:
# {"physical_file_id": ..., "document_id": ...}
#
# That could change ordering when multiple utterances had
# identical token counts and produced a false audit failure.
#
# NO target replacement.
# NO pilot replacement.
# NO question generation.
# NO expected-answer generation.
# NO artifact writes.
# ============================================================


# ============================================================
# 1) Identify the two affected selected strata
# ============================================================

exercise_selected_cases = [
    case_record
    for case_record
    in golden_case_records
    if (
        case_record[
            "source_type"
        ]
        == "book"

        and

        case_record[
            "stratum"
        ]
        == "exercise_or_self_assessment"

        and

        case_record[
            "final_golden_membership"
        ]
        is True
    )
]


single_primary_selected_cases = [
    case_record
    for case_record
    in golden_case_records
    if (
        case_record[
            "source_type"
        ]
        == "transcript"

        and

        case_record[
            "stratum"
        ]
        == "single_speaker_primary"

        and

        case_record[
            "final_golden_membership"
        ]
        is True
    )
]


# ============================================================
# 2) Build Book records grouped by exact structural heading path
# ============================================================

book_heading_groups = defaultdict(
    list
)


for document_id, record in (
    corpus_target_index.items()
):

    if (
        record[
            "source_type"
        ]
        != "book"
    ):
        continue

    locator = (
        record[
            "locator"
        ]
        or {}
    )

    metadata = (
        record[
            "source_metadata"
        ]
        or {}
    )

    edition_id = (
        locator.get(
            "edition_id"
        )
    )

    work_id = (
        locator.get(
            "work_id"
        )
    )

    heading_path = (
        locator.get(
            "heading_path"
        )
    )

    if (
        heading_path
        is None
    ):

        heading_path = (
            metadata.get(
                "heading_path"
            )
        )

    if isinstance(
        heading_path,
        list,
    ):

        heading_path_key = tuple(
            str(
                value
            )
            for value
            in heading_path
        )

    elif (
        heading_path
        is None
    ):

        heading_path_key = tuple()

    else:

        heading_path_key = (
            str(
                heading_path
            ),
        )


    group_key = (
        work_id,
        edition_id,
        heading_path_key,
    )


    book_heading_groups[
        group_key
    ].append(
        {
            "document_id":
                document_id,

            "paragraph_number":
                (
                    locator.get(
                        "paragraph_number"
                    )
                    if (
                        locator.get(
                            "paragraph_number"
                        )
                        is not None
                    )
                    else
                    metadata.get(
                        "paragraph_number"
                    )
                ),

            "is_heading":
                metadata.get(
                    "is_heading"
                ),

            "in_table":
                metadata.get(
                    "in_table"
                ),

            "book_unit_type":
                metadata.get(
                    "book_unit_type"
                ),

            "text":
                record[
                    "text"
                ],

            "token_count":
                nb11_token_count(
                    record[
                        "text"
                    ]
                ),
        }
    )


for group_key in list(
    book_heading_groups
):

    book_heading_groups[
        group_key
    ] = sorted(
        book_heading_groups[
            group_key
        ],
        key=lambda item: (
            item[
                "paragraph_number"
            ]
            if (
                item[
                    "paragraph_number"
                ]
                is not None
            )
            else 10**12,
            item[
                "document_id"
            ],
        ),
    )


# ============================================================
# 3) Audit all selected exercise/self-assessment targets
# ============================================================

exercise_method_audit_rows = []


for case_record in (
    exercise_selected_cases
):

    candidate = (
        case_record[
            "candidate_snapshot"
        ]
    )

    document_id = (
        candidate[
            "document_id"
        ]
    )

    source_record = (
        corpus_target_index[
            document_id
        ]
    )

    locator = (
        source_record[
            "locator"
        ]
        or {}
    )

    metadata = (
        source_record[
            "source_metadata"
        ]
        or {}
    )

    heading_path = (
        locator.get(
            "heading_path"
        )
    )

    if (
        heading_path
        is None
    ):

        heading_path = (
            metadata.get(
                "heading_path"
            )
        )

    if isinstance(
        heading_path,
        list,
    ):

        heading_path_key = tuple(
            str(
                value
            )
            for value
            in heading_path
        )

    elif (
        heading_path
        is None
    ):

        heading_path_key = tuple()

    else:

        heading_path_key = (
            str(
                heading_path
            ),
        )


    group_key = (
        locator.get(
            "work_id"
        ),
        locator.get(
            "edition_id"
        ),
        heading_path_key,
    )


    group_records = (
        book_heading_groups.get(
            group_key,
            [],
        )
    )


    candidate_position = None


    for index, group_record in enumerate(
        group_records
    ):

        if (
            group_record[
                "document_id"
            ]
            ==
            document_id
        ):

            candidate_position = index
            break


    assert (
        candidate_position
        is not None
    )


    candidate_tokens = (
        nb11_token_count(
            source_record[
                "text"
            ]
        )
    )


    group_total_tokens = sum(
        record[
            "token_count"
        ]
        for record
        in group_records
    )


    group_nonheading_tokens = sum(
        record[
            "token_count"
        ]
        for record
        in group_records
        if (
            record[
                "is_heading"
            ]
            is not True
        )
    )


    immediate_neighbor_records = []

    for neighbor_index in [
        candidate_position - 2,
        candidate_position - 1,
        candidate_position + 1,
        candidate_position + 2,
    ]:

        if (
            0
            <= neighbor_index
            < len(
                group_records
            )
        ):

            immediate_neighbor_records.append(
                group_records[
                    neighbor_index
                ]
            )


    exercise_method_audit_rows.append(
        {
            "golden_case_id":
                case_record[
                    "golden_case_id"
                ],

            "document_id":
                document_id,

            "heading_path":
                list(
                    heading_path_key
                ),

            "candidate_tokens":
                candidate_tokens,

            "group_record_count":
                len(
                    group_records
                ),

            "group_total_tokens":
                group_total_tokens,

            "group_nonheading_tokens":
                group_nonheading_tokens,

            "candidate_share_of_group_tokens":
                (
                    candidate_tokens
                    /
                    group_total_tokens
                    if (
                        group_total_tokens
                        > 0
                    )
                    else None
                ),

            "candidate_position":
                candidate_position,

            "candidate_text":
                source_record[
                    "text"
                ],

            "immediate_neighbor_records":
                [
                    {
                        "document_id":
                            record[
                                "document_id"
                            ],

                        "paragraph_number":
                            record[
                                "paragraph_number"
                            ],

                        "is_heading":
                            record[
                                "is_heading"
                            ],

                        "in_table":
                            record[
                                "in_table"
                            ],

                        "token_count":
                            record[
                                "token_count"
                            ],

                        "text":
                            record[
                                "text"
                            ],
                    }

                    for record
                    in immediate_neighbor_records
                ],

            "full_heading_group_preview":
                [
                    {
                        "document_id":
                            record[
                                "document_id"
                            ],

                        "paragraph_number":
                            record[
                                "paragraph_number"
                            ],

                        "is_heading":
                            record[
                                "is_heading"
                            ],

                        "in_table":
                            record[
                                "in_table"
                            ],

                        "token_count":
                            record[
                                "token_count"
                            ],

                        "text_preview":
                            record[
                                "text"
                            ][
                                :300
                            ],
                    }

                    for record
                    in group_records[
                        :12
                    ]
                ],
        }
    )


# ============================================================
# 4) Exercise structural distributions
# ============================================================

exercise_group_size_distribution = Counter(
    row[
        "group_record_count"
    ]
    for row
    in exercise_method_audit_rows
)


exercise_singleton_heading_group_count = sum(
    1
    for row
    in exercise_method_audit_rows
    if (
        row[
            "group_record_count"
        ]
        == 1
    )
)


exercise_has_additional_same_heading_evidence_count = sum(
    1
    for row
    in exercise_method_audit_rows
    if (
        row[
            "group_record_count"
        ]
        > 1
    )
)


exercise_candidate_token_values = [
    row[
        "candidate_tokens"
    ]
    for row
    in exercise_method_audit_rows
]


exercise_group_token_values = [
    row[
        "group_total_tokens"
    ]
    for row
    in exercise_method_audit_rows
]


exercise_candidate_share_values = [
    row[
        "candidate_share_of_group_tokens"
    ]
    for row
    in exercise_method_audit_rows
    if (
        row[
            "candidate_share_of_group_tokens"
        ]
        is not None
    )
]


# ============================================================
# 5) Locate failed exercise Pilot specifically
# ============================================================

failed_exercise_pilot = next(
    row
    for row
    in pilot_semantic_review_rows
    if (
        row[
            "source_type"
        ]
        == "book"

        and

        row[
            "stratum"
        ]
        == "exercise_or_self_assessment"
    )
)


failed_exercise_case_id = (
    failed_exercise_pilot[
        "golden_case_id"
    ]
)


failed_exercise_audit = next(
    row
    for row
    in exercise_method_audit_rows
    if (
        row[
            "golden_case_id"
        ]
        ==
        failed_exercise_case_id
    )
)


# ============================================================
# 6) Audit all single-speaker-primary physical files
#
# IMPORTANT:
# Reproduce Cell 28 EXACTLY.
#
# Cell 28 used:
#   stable_target_tie_hash(
#       {"document_id": record["document_id"]}
#   )
#
# No physical_file_id is allowed in this audit tie-break.
# ============================================================

single_primary_method_audit_rows = []

single_primary_exact_reproduction_failures = []

single_primary_equal_max_token_ties = []


for case_record in (
    single_primary_selected_cases
):

    candidate = (
        case_record[
            "candidate_snapshot"
        ]
    )

    physical_file_id = (
        candidate[
            "physical_file_id"
        ]
    )

    records = (
        transcript_records_by_physical_file[
            physical_file_id
        ]
    )


    assert records


    ranked_utterances = []


    for record in (
        records
    ):

        token_count = (
            nb11_token_count(
                record[
                    "text"
                ]
            )
        )


        # ----------------------------------------------------
        # EXACT Cell 28 tie-break reproduction
        # ----------------------------------------------------

        tie_hash = (
            stable_target_tie_hash(
                {
                    "document_id":
                        record[
                            "document_id"
                        ]
                }
            )
        )


        ranked_utterances.append(
            {
                "document_id":
                    record[
                        "document_id"
                    ],

                "utterance_index":
                    record[
                        "utterance_index"
                    ],

                "token_count":
                    token_count,

                "stable_tie_hash":
                    tie_hash,

                "speaker_signature":
                    transcript_speaker_signature(
                        record
                    ),

                "text":
                    record[
                        "text"
                    ],
            }
        )


    ranked_utterances = sorted(
        ranked_utterances,
        key=lambda item: (
            -item[
                "token_count"
            ],
            item[
                "stable_tie_hash"
            ],
        ),
    )


    expected_winner = (
        ranked_utterances[
            0
        ]
    )


    current_target = (
        target_evidence_by_case_id[
            case_record[
                "golden_case_id"
            ]
        ]
    )


    current_document_id = (
        current_target[
            "target_document_ids"
        ][
            0
        ]
    )


    current_rank = next(
        index
        for index, utterance
        in enumerate(
            ranked_utterances,
            start=1,
        )
        if (
            utterance[
                "document_id"
            ]
            ==
            current_document_id
        )
    )


    exact_reproduction_match = (
        current_document_id
        ==
        expected_winner[
            "document_id"
        ]
    )


    if not exact_reproduction_match:

        single_primary_exact_reproduction_failures.append(
            {
                "golden_case_id":
                    case_record[
                        "golden_case_id"
                    ],

                "physical_file_id":
                    physical_file_id,

                "current_target_document_id":
                    current_document_id,

                "expected_document_id":
                    expected_winner[
                        "document_id"
                    ],

                "current_rank":
                    current_rank,

                "expected_tokens":
                    expected_winner[
                        "token_count"
                    ],
            }
        )


    max_token_count = (
        expected_winner[
            "token_count"
        ]
    )


    equal_max_candidates = [
        utterance
        for utterance
        in ranked_utterances
        if (
            utterance[
                "token_count"
            ]
            ==
            max_token_count
        )
    ]


    if (
        len(
            equal_max_candidates
        )
        > 1
    ):

        single_primary_equal_max_token_ties.append(
            {
                "golden_case_id":
                    case_record[
                        "golden_case_id"
                    ],

                "physical_file_id":
                    physical_file_id,

                "max_token_count":
                    max_token_count,

                "tied_document_ids":
                    [
                        utterance[
                            "document_id"
                        ]
                        for utterance
                        in equal_max_candidates
                    ],

                "winner_after_cell28_tie_break":
                    expected_winner[
                        "document_id"
                    ],
            }
        )


    single_primary_method_audit_rows.append(
        {
            "golden_case_id":
                case_record[
                    "golden_case_id"
                ],

            "physical_file_id":
                physical_file_id,

            "program_id":
                candidate.get(
                    "program_id"
                ),

            "episode_number":
                candidate.get(
                    "episode_number"
                ),

            "utterance_count":
                len(
                    records
                ),

            "current_target_document_id":
                current_document_id,

            "expected_cell28_target_document_id":
                expected_winner[
                    "document_id"
                ],

            "exact_cell28_reproduction_match":
                exact_reproduction_match,

            "current_target_rank_by_exact_cell28_order":
                current_rank,

            "current_target_tokens":
                next(
                    utterance[
                        "token_count"
                    ]
                    for utterance
                    in ranked_utterances
                    if (
                        utterance[
                            "document_id"
                        ]
                        ==
                        current_document_id
                    )
                ),

            "max_token_tie_count":
                len(
                    equal_max_candidates
                ),

            "alternative_utterance_count":
                max(
                    0,
                    len(
                        ranked_utterances
                    )
                    - 1,
                ),

            "top_utterance_candidates":
                [
                    {
                        "rank":
                            rank,

                        "document_id":
                            utterance[
                                "document_id"
                            ],

                        "utterance_index":
                            utterance[
                                "utterance_index"
                            ],

                        "token_count":
                            utterance[
                                "token_count"
                            ],

                        "stable_tie_hash":
                            utterance[
                                "stable_tie_hash"
                            ],

                        "speaker_signature":
                            utterance[
                                "speaker_signature"
                            ],

                        "text_preview":
                            utterance[
                                "text"
                            ][
                                :500
                            ],
                    }

                    for rank, utterance
                    in enumerate(
                        ranked_utterances[
                            :8
                        ],
                        start=1,
                    )
                ],
        }
    )


# ============================================================
# 7) Locate failed transcript Pilot specifically
# ============================================================

failed_single_primary_pilot = next(
    row
    for row
    in pilot_semantic_review_rows
    if (
        row[
            "source_type"
        ]
        == "transcript"

        and

        row[
            "stratum"
        ]
        == "single_speaker_primary"
    )
)


failed_single_primary_case_id = (
    failed_single_primary_pilot[
        "golden_case_id"
    ]
)


failed_single_primary_audit = next(
    row
    for row
    in single_primary_method_audit_rows
    if (
        row[
            "golden_case_id"
        ]
        ==
        failed_single_primary_case_id
    )
)


# ============================================================
# 8) Read-only discovery of possible persisted semantic signals
#
# Search only semantic-looking JSON / JSONL artifacts.
#
# Discovery only.
# Absence here is NOT proof of global absence.
# ============================================================

repo_root = (
    corpus_path.parents[
        3
    ]
)


semantic_search_roots = [
    repo_root
    /
    "artifacts"
    /
    "phase2",

    repo_root
    /
    "data"
    /
    "phase2"
    /
    "processed",

    repo_root
    /
    "reports"
    /
    "phase2",
]


semantic_filename_markers = [
    "taxonomy",
    "topic",
    "competency",
    "mapping",
    "knowledge",
]


semantic_candidate_paths = []


for search_root in (
    semantic_search_roots
):

    if not search_root.exists():
        continue

    for path in (
        search_root.rglob(
            "*"
        )
    ):

        if (
            not path.is_file()
        ):
            continue

        if (
            path.suffix.lower()
            not in {
                ".json",
                ".jsonl",
            }
        ):
            continue

        name_lower = (
            path.name.lower()
        )

        if any(
            marker
            in name_lower
            for marker
            in semantic_filename_markers
        ):

            semantic_candidate_paths.append(
                path
            )


semantic_candidate_paths = sorted(
    set(
        semantic_candidate_paths
    )
)


semantic_artifact_discovery_rows = []


for path in (
    semantic_candidate_paths
):

    with path.open(
        "r",
        encoding="utf-8",
        errors="replace",
    ) as handle:

        sample_text = (
            handle.read(
                200000
            )
        )


    keyword_hits = [
        keyword
        for keyword
        in [
            '"document_id"',
            '"source_id"',
            '"topic"',
            '"topics"',
            '"taxonomy"',
            '"competency"',
            '"mapping"',
        ]
        if (
            keyword
            in sample_text
        )
    ]


    semantic_artifact_discovery_rows.append(
        {
            "path":
                str(
                    path.relative_to(
                        repo_root
                    )
                ),

            "size_bytes":
                path.stat().st_size,

            "sample_contains_document_id":
                (
                    '"document_id"'
                    in sample_text
                ),

            "sample_contains_source_id":
                (
                    '"source_id"'
                    in sample_text
                ),

            "semantic_keyword_hits":
                keyword_hits,
        }
    )


document_level_semantic_signal_candidates = [
    row
    for row
    in semantic_artifact_discovery_rows
    if (
        row[
            "sample_contains_document_id"
        ]

        and

        any(
            keyword
            in row[
                "semantic_keyword_hits"
            ]
            for keyword
            in [
                '"topic"',
                '"topics"',
                '"taxonomy"',
                '"competency"',
                '"mapping"',
            ]
        )
    )
]


# ============================================================
# 9) Descriptive repair-scope evidence
# ============================================================

single_primary_utterance_count_values = [
    row[
        "utterance_count"
    ]
    for row
    in single_primary_method_audit_rows
]


single_primary_with_alternatives_count = sum(
    1
    for row
    in single_primary_method_audit_rows
    if (
        row[
            "alternative_utterance_count"
        ]
        > 0
    )
)


single_primary_exact_reproduction_match_count = sum(
    1
    for row
    in single_primary_method_audit_rows
    if (
        row[
            "exact_cell28_reproduction_match"
        ]
        is True
    )
)


single_primary_max_token_tie_case_count = (
    len(
        single_primary_equal_max_token_ties
    )
)


# ============================================================
# 10) Evidence-derived repair-scope interpretation
#
# These are audit findings only.
# No target changes happen here.
# ============================================================

exercise_method_systemic_structure_signal = (
    len(
        exercise_method_audit_rows
    )
    > 0

    and

    exercise_has_additional_same_heading_evidence_count
    ==
    len(
        exercise_method_audit_rows
    )
)


single_primary_method_exactly_reproduced = (
    len(
        single_primary_exact_reproduction_failures
    )
    == 0
)


# ============================================================
# 11) Repair Scope Audit Registry
# ============================================================

REPAIR_SCOPE_AUDIT_V1_DRAFT = {
    "version":
        "repair_scope_audit_v1_draft",

    "status":
        "INSPECTION_COMPLETE_NO_REPAIR_EXECUTED",

    "exercise_selected_case_count":
        len(
            exercise_method_audit_rows
        ),

    "single_primary_selected_case_count":
        len(
            single_primary_method_audit_rows
        ),

    "exercise_cases_with_additional_same_heading_evidence":
        exercise_has_additional_same_heading_evidence_count,

    "exercise_singleton_heading_groups":
        exercise_singleton_heading_group_count,

    "exercise_method_systemic_structure_signal":
        exercise_method_systemic_structure_signal,

    "single_primary_cases_with_alternative_utterances":
        single_primary_with_alternatives_count,

    "single_primary_exact_cell28_reproduction_matches":
        single_primary_exact_reproduction_match_count,

    "single_primary_max_token_tie_case_count":
        single_primary_max_token_tie_case_count,

    "semantic_artifact_candidates_discovered":
        len(
            semantic_artifact_discovery_rows
        ),

    "possible_document_level_semantic_signal_artifacts":
        len(
            document_level_semantic_signal_candidates
        ),

    "semantic_discovery_limitation":
        (
            "FILENAME_AND_BOUNDED_CONTENT_DISCOVERY_ONLY_"
            "NOT_PROOF_OF_GLOBAL_ABSENCE"
        ),

    "repair_execution_status":
        "NOT_STARTED",

    "question_generation_status":
        "NOT_STARTED",

    "expected_answer_generation_status":
        "NOT_STARTED",

    "artifact_write_status":
        "NONE",
}


# ============================================================
# 12) Validation gates
# ============================================================

exact_38_exercise_cases_audited_gate = (
    len(
        exercise_method_audit_rows
    )
    == 38
)


exact_38_single_primary_cases_audited_gate = (
    len(
        single_primary_method_audit_rows
    )
    == 38
)


failed_exercise_pilot_resolved_in_audit_gate = (
    failed_exercise_audit[
        "golden_case_id"
    ]
    ==
    failed_exercise_case_id
)


failed_single_primary_pilot_resolved_in_audit_gate = (
    failed_single_primary_audit[
        "golden_case_id"
    ]
    ==
    failed_single_primary_case_id
)


exercise_heading_group_resolution_gate = all(
    row[
        "group_record_count"
    ]
    >= 1

    and

    row[
        "group_total_tokens"
    ]
    >=
    row[
        "candidate_tokens"
    ]

    for row
    in exercise_method_audit_rows
)


# Exact reproduction of Cell 28 is the correct gate.
single_primary_current_method_reproduced_gate = (
    len(
        single_primary_exact_reproduction_failures
    )
    == 0

    and

    single_primary_exact_reproduction_match_count
    ==
    38
)


all_single_primary_files_have_records_gate = all(
    row[
        "utterance_count"
    ]
    >= 1

    for row
    in single_primary_method_audit_rows
)


repair_not_executed_gate = (
    REPAIR_SCOPE_AUDIT_V1_DRAFT[
        "repair_execution_status"
    ]
    ==
    "NOT_STARTED"
)


question_generation_not_started_gate = (
    REPAIR_SCOPE_AUDIT_V1_DRAFT[
        "question_generation_status"
    ]
    ==
    "NOT_STARTED"
)


expected_answer_not_started_gate = (
    REPAIR_SCOPE_AUDIT_V1_DRAFT[
        "expected_answer_generation_status"
    ]
    ==
    "NOT_STARTED"
)


no_artifact_write_gate = (
    REPAIR_SCOPE_AUDIT_V1_DRAFT[
        "artifact_write_status"
    ]
    ==
    "NONE"
)


cell_32_gates = {
    "exact_38_exercise_cases_audited":
        exact_38_exercise_cases_audited_gate,

    "exact_38_single_speaker_primary_cases_audited":
        exact_38_single_primary_cases_audited_gate,

    "failed_exercise_pilot_resolved_in_scope_audit":
        failed_exercise_pilot_resolved_in_audit_gate,

    "failed_single_primary_pilot_resolved_in_scope_audit":
        failed_single_primary_pilot_resolved_in_audit_gate,

    "all_exercise_targets_resolve_same_heading_structure":
        exercise_heading_group_resolution_gate,

    "current_single_primary_max_token_method_reproduced":
        single_primary_current_method_reproduced_gate,

    "all_single_primary_files_have_real_utterances":
        all_single_primary_files_have_records_gate,

    "repair_execution_not_started":
        repair_not_executed_gate,

    "question_generation_not_started":
        question_generation_not_started_gate,

    "expected_answer_generation_not_started":
        expected_answer_not_started_gate,

    "no_artifact_write":
        no_artifact_write_gate,
}


cell_32_pass = all(
    cell_32_gates.values()
)


# ============================================================
# 13) Output
# ============================================================

print(
    "=== SYSTEMIC REPAIR SCOPE AUDIT ==="
)


print(
    "\n=== BOOK / EXERCISE_OR_SELF_ASSESSMENT ==="
)

print(
    "Selected cases audited:",
    len(
        exercise_method_audit_rows
    )
)

print(
    "Cases with additional same-heading evidence:",
    exercise_has_additional_same_heading_evidence_count
)

print(
    "Singleton same-heading groups:",
    exercise_singleton_heading_group_count
)

print(
    "Systemic same-heading structure signal:",
    exercise_method_systemic_structure_signal
)


print(
    "Candidate target token distribution:",
    {
        "min":
            min(
                exercise_candidate_token_values
            ),

        "median":
            empirical_nearest_rank(
                exercise_candidate_token_values,
                0.50,
            ),

        "max":
            max(
                exercise_candidate_token_values
            ),
    }
)


print(
    "Full same-heading token distribution:",
    {
        "min":
            min(
                exercise_group_token_values
            ),

        "median":
            empirical_nearest_rank(
                exercise_group_token_values,
                0.50,
            ),

        "max":
            max(
                exercise_group_token_values
            ),
    }
)


print(
    "Candidate share of same-heading token distribution:",
    {
        "min":
            min(
                exercise_candidate_share_values
            ),

        "median":
            empirical_nearest_rank(
                exercise_candidate_share_values,
                0.50,
            ),

        "max":
            max(
                exercise_candidate_share_values
            ),
    }
)


print(
    "\nFailed Pilot structural context:"
)

print(
    json.dumps(
        failed_exercise_audit,
        ensure_ascii=False,
        sort_keys=True,
    )
)


print(
    "\n=== TRANSCRIPT / SINGLE_SPEAKER_PRIMARY ==="
)

print(
    "Selected cases audited:",
    len(
        single_primary_method_audit_rows
    )
)

print(
    "Cases with alternative utterances:",
    single_primary_with_alternatives_count
)

print(
    "Exact Cell 28 reproduction matches:",
    single_primary_exact_reproduction_match_count,
    "/",
    len(
        single_primary_method_audit_rows
    )
)

print(
    "Exact reproduction failures:",
    len(
        single_primary_exact_reproduction_failures
    )
)

print(
    "Cases with tied maximum token count:",
    single_primary_max_token_tie_case_count
)


print(
    "Utterance-count distribution:",
    {
        "min":
            min(
                single_primary_utterance_count_values
            ),

        "median":
            empirical_nearest_rank(
                single_primary_utterance_count_values,
                0.50,
            ),

        "max":
            max(
                single_primary_utterance_count_values
            ),
    }
)


print(
    "\nFailed Pilot top deterministic alternatives:"
)

print(
    json.dumps(
        failed_single_primary_audit,
        ensure_ascii=False,
        sort_keys=True,
    )
)


if (
    single_primary_equal_max_token_ties
):

    print(
        "\n=== MAX-TOKEN TIE CASES ==="
    )

    for row in (
        single_primary_equal_max_token_ties
    ):

        print(
            json.dumps(
                row,
                ensure_ascii=False,
                sort_keys=True,
            )
        )


if (
    single_primary_exact_reproduction_failures
):

    print(
        "\n=== EXACT REPRODUCTION FAILURES ==="
    )

    for row in (
        single_primary_exact_reproduction_failures
    ):

        print(
            json.dumps(
                row,
                ensure_ascii=False,
                sort_keys=True,
            )
        )


print(
    "\n=== POSSIBLE PERSISTED SEMANTIC SIGNALS ==="
)

print(
    "Semantic-looking JSON/JSONL artifacts discovered:",
    len(
        semantic_artifact_discovery_rows
    )
)

print(
    "Possible document-level semantic signal artifacts:",
    len(
        document_level_semantic_signal_candidates
    )
)


for row in (
    semantic_artifact_discovery_rows
):

    print(
        "-",
        row[
            "path"
        ],
        "| bytes:",
        row[
            "size_bytes"
        ],
        "| document_id:",
        row[
            "sample_contains_document_id"
        ],
        "| source_id:",
        row[
            "sample_contains_source_id"
        ],
        "| hits:",
        row[
            "semantic_keyword_hits"
        ],
    )


print(
    "\n=== POSSIBLE DOCUMENT-LEVEL SIGNAL CANDIDATES ==="
)

for row in (
    document_level_semantic_signal_candidates
):

    print(
        "-",
        row[
            "path"
        ]
    )


print(
    "\nDiscovery limitation:",
    REPAIR_SCOPE_AUDIT_V1_DRAFT[
        "semantic_discovery_limitation"
    ]
)


print(
    "\n=== CODE CELL 32 GATES ==="
)

for name, passed in (
    cell_32_gates.items()
):

    print(
        name,
        ":",
        "PASS"
        if passed
        else "FAIL"
    )


print(
    "\nCODE CELL 32 GATE:",
    "PASS"
    if cell_32_pass
    else "FAIL"
)


assert cell_32_pass, (
    "Systemic repair-scope audit failed. "
    "Do not modify either Target Selection Method until the "
    "38 Book cases and 38 single-speaker-primary transcript "
    "cases are structurally inspected and the Cell 28 "
    "deterministic method is reproduced exactly."
)
```

```python
# Code Cell 33

# ============================================================
# Repair Feasibility & Evidence-Scope Audit
#
# INSPECTION ONLY.
#
# NO Golden Case mutation.
# NO Target Evidence mutation.
# NO Pilot replacement.
# NO question generation.
# NO expected-answer generation.
# NO artifact writes.
# ============================================================


# ============================================================
# 1) BOOK — compare exact same-heading group with current
# Golden Case evidence scope for all 38 exercise cases
# ============================================================

exercise_evidence_scope_audit_rows = []


for audit_row in (
    exercise_method_audit_rows
):

    case_id = (
        audit_row[
            "golden_case_id"
        ]
    )

    case_record = (
        golden_case_records_by_id[
            case_id
        ]
    )

    current_case_evidence_ids = set(
        case_record[
            "evidence_document_ids"
        ]
    )


    candidate_document_id = (
        audit_row[
            "document_id"
        ]
    )

    candidate_record = (
        corpus_target_index[
            candidate_document_id
        ]
    )

    locator = (
        candidate_record[
            "locator"
        ]
        or {}
    )

    metadata = (
        candidate_record[
            "source_metadata"
        ]
        or {}
    )


    heading_path = (
        locator.get(
            "heading_path"
        )
    )

    if (
        heading_path
        is None
    ):

        heading_path = (
            metadata.get(
                "heading_path"
            )
        )


    if isinstance(
        heading_path,
        list,
    ):

        heading_path_key = tuple(
            str(
                value
            )
            for value
            in heading_path
        )

    elif (
        heading_path
        is None
    ):

        heading_path_key = tuple()

    else:

        heading_path_key = (
            str(
                heading_path
            ),
        )


    group_key = (
        locator.get(
            "work_id"
        ),
        locator.get(
            "edition_id"
        ),
        heading_path_key,
    )


    same_heading_records = (
        book_heading_groups[
            group_key
        ]
    )


    same_heading_document_ids = [
        record[
            "document_id"
        ]
        for record
        in same_heading_records
    ]


    same_heading_document_id_set = set(
        same_heading_document_ids
    )


    already_in_case_evidence = sorted(
        same_heading_document_id_set
        &
        current_case_evidence_ids
    )


    missing_from_case_evidence = [
        document_id
        for document_id
        in same_heading_document_ids
        if (
            document_id
            not in
            current_case_evidence_ids
        )
    ]


    extra_case_evidence_outside_heading = sorted(
        current_case_evidence_ids
        -
        same_heading_document_id_set
    )


    same_heading_fully_inside_case_evidence = (
        len(
            missing_from_case_evidence
        )
        == 0
    )


    exercise_evidence_scope_audit_rows.append(
        {
            "golden_case_id":
                case_id,

            "candidate_document_id":
                candidate_document_id,

            "heading_path":
                list(
                    heading_path_key
                ),

            "current_case_evidence_count":
                len(
                    current_case_evidence_ids
                ),

            "same_heading_record_count":
                len(
                    same_heading_document_ids
                ),

            "same_heading_inside_current_case_count":
                len(
                    already_in_case_evidence
                ),

            "same_heading_missing_from_current_case_count":
                len(
                    missing_from_case_evidence
                ),

            "same_heading_fully_inside_case_evidence":
                same_heading_fully_inside_case_evidence,

            "missing_from_case_evidence":
                missing_from_case_evidence,

            "extra_case_evidence_outside_heading":
                extra_case_evidence_outside_heading,
        }
    )


# ============================================================
# 2) Book evidence-scope summary
# ============================================================

exercise_full_heading_already_inside_case_count = sum(
    1
    for row
    in exercise_evidence_scope_audit_rows
    if (
        row[
            "same_heading_fully_inside_case_evidence"
        ]
        is True
    )
)


exercise_requires_case_scope_expansion_count = sum(
    1
    for row
    in exercise_evidence_scope_audit_rows
    if (
        row[
            "same_heading_fully_inside_case_evidence"
        ]
        is False
    )
)


exercise_missing_heading_records_total = sum(
    row[
        "same_heading_missing_from_current_case_count"
    ]
    for row
    in exercise_evidence_scope_audit_rows
)


failed_exercise_scope_audit = next(
    row
    for row
    in exercise_evidence_scope_audit_rows
    if (
        row[
            "golden_case_id"
        ]
        ==
        failed_exercise_case_id
    )
)


# ============================================================
# 3) Independently validate every proposed same-heading
# document remains in the same Book Work + Edition + Heading
# Path as the candidate.
# ============================================================

exercise_same_structure_failures = []


for audit_row in (
    exercise_evidence_scope_audit_rows
):

    candidate_record = (
        corpus_target_index[
            audit_row[
                "candidate_document_id"
            ]
        ]
    )

    candidate_locator = (
        candidate_record[
            "locator"
        ]
        or {}
    )

    candidate_metadata = (
        candidate_record[
            "source_metadata"
        ]
        or {}
    )


    candidate_heading_path = (
        candidate_locator.get(
            "heading_path"
        )
    )

    if (
        candidate_heading_path
        is None
    ):

        candidate_heading_path = (
            candidate_metadata.get(
                "heading_path"
            )
        )


    candidate_heading_key = tuple(
        candidate_heading_path
        or []
    )


    for document_id in (
        audit_row[
            "missing_from_case_evidence"
        ]
    ):

        record = (
            corpus_target_index[
                document_id
            ]
        )

        locator = (
            record[
                "locator"
            ]
            or {}
        )

        metadata = (
            record[
                "source_metadata"
            ]
            or {}
        )


        heading_path = (
            locator.get(
                "heading_path"
            )
        )

        if (
            heading_path
            is None
        ):

            heading_path = (
                metadata.get(
                    "heading_path"
                )
            )


        heading_key = tuple(
            heading_path
            or []
        )


        same_structure = (
            locator.get(
                "work_id"
            )
            ==
            candidate_locator.get(
                "work_id"
            )

            and

            locator.get(
                "edition_id"
            )
            ==
            candidate_locator.get(
                "edition_id"
            )

            and

            heading_key
            ==
            candidate_heading_key
        )


        if not same_structure:

            exercise_same_structure_failures.append(
                {
                    "golden_case_id":
                        audit_row[
                            "golden_case_id"
                        ],

                    "candidate_document_id":
                        audit_row[
                            "candidate_document_id"
                        ],

                    "proposed_document_id":
                        document_id,
                }
            )


# ============================================================
# 4) TRANSCRIPT — verify every alternative utterance in the
# failed Pilot remains inside current Golden Case evidence.
# ============================================================

failed_single_primary_case_record = (
    golden_case_records_by_id[
        failed_single_primary_case_id
    ]
)


failed_single_primary_case_evidence_ids = set(
    failed_single_primary_case_record[
        "evidence_document_ids"
    ]
)


failed_single_primary_physical_file_id = (
    failed_single_primary_case_record[
        "candidate_snapshot"
    ][
        "physical_file_id"
    ]
)


failed_single_primary_file_records = (
    transcript_records_by_physical_file[
        failed_single_primary_physical_file_id
    ]
)


failed_single_primary_file_document_ids = {
    record[
        "document_id"
    ]
    for record
    in failed_single_primary_file_records
}


transcript_file_fully_inside_case_evidence = (
    failed_single_primary_file_document_ids
    .issubset(
        failed_single_primary_case_evidence_ids
    )
)


transcript_file_missing_from_case_evidence = sorted(
    failed_single_primary_file_document_ids
    -
    failed_single_primary_case_evidence_ids
)


transcript_case_extra_outside_file = sorted(
    failed_single_primary_case_evidence_ids
    -
    failed_single_primary_file_document_ids
)


# ============================================================
# 5) Inspect exact candidate alternatives already surfaced
# in Cell 32 and mark whether they are scope-eligible.
#
# NO semantic winner is selected here.
# ============================================================

failed_single_primary_alternative_scope_rows = []


for candidate_row in (
    failed_single_primary_audit[
        "top_utterance_candidates"
    ]
):

    document_id = (
        candidate_row[
            "document_id"
        ]
    )


    failed_single_primary_alternative_scope_rows.append(
        {
            "rank":
                candidate_row[
                    "rank"
                ],

            "document_id":
                document_id,

            "utterance_index":
                candidate_row[
                    "utterance_index"
                ],

            "token_count":
                candidate_row[
                    "token_count"
                ],

            "inside_current_golden_case_evidence":
                (
                    document_id
                    in
                    failed_single_primary_case_evidence_ids
                ),

            "inside_same_physical_file":
                (
                    document_id
                    in
                    failed_single_primary_file_document_ids
                ),

            "text_preview":
                candidate_row[
                    "text_preview"
                ],
        }
    )


# ============================================================
# 6) Check whether candidate records have complete lineage /
# locator / attribution required for a potential repair.
# ============================================================

repair_candidate_integrity_failures = []


# Book proposed evidence expansion records
for row in (
    exercise_evidence_scope_audit_rows
):

    for document_id in (
        row[
            "missing_from_case_evidence"
        ]
    ):

        record = (
            corpus_target_index[
                document_id
            ]
        )

        if not (
            record[
                "locator"
            ]
        ):

            repair_candidate_integrity_failures.append(
                (
                    document_id,
                    "MISSING_LOCATOR",
                )
            )

        if (
            record[
                "lineage"
            ]
            is None
        ):

            repair_candidate_integrity_failures.append(
                (
                    document_id,
                    "MISSING_LINEAGE",
                )
            )

        if (
            record[
                "attribution"
            ]
            is None
        ):

            repair_candidate_integrity_failures.append(
                (
                    document_id,
                    "MISSING_ATTRIBUTION",
                )
            )

        if (
            record[
                "rights"
            ]
            is None
        ):

            repair_candidate_integrity_failures.append(
                (
                    document_id,
                    "MISSING_RIGHTS",
                )
            )


# Transcript alternatives
for row in (
    failed_single_primary_alternative_scope_rows
):

    document_id = (
        row[
            "document_id"
        ]
    )

    record = (
        corpus_target_index[
            document_id
        ]
    )

    if not (
        record[
            "locator"
        ]
    ):

        repair_candidate_integrity_failures.append(
            (
                document_id,
                "MISSING_LOCATOR",
            )
        )

    if (
        record[
            "lineage"
        ]
        is None
    ):

        repair_candidate_integrity_failures.append(
            (
                document_id,
                "MISSING_LINEAGE",
            )
        )

    if (
        record[
            "attribution"
        ]
        is None
    ):

        repair_candidate_integrity_failures.append(
            (
                document_id,
                "MISSING_ATTRIBUTION",
            )
        )

    if (
        record[
            "rights"
        ]
        is None
    ):

        repair_candidate_integrity_failures.append(
            (
                document_id,
                "MISSING_RIGHTS",
            )
        )


# ============================================================
# 7) Evidence-derived repair feasibility classification
#
# This does not execute the repair.
# ============================================================

if (
    exercise_requires_case_scope_expansion_count
    > 0
):

    exercise_repair_classification = (
        "CASE_EVIDENCE_SCOPE_REPAIR_REQUIRED_BEFORE_TARGET_REPAIR"
    )

else:

    exercise_repair_classification = (
        "TARGET_METHOD_REPAIR_ONLY_IS_FEASIBLE"
    )


if (
    transcript_file_fully_inside_case_evidence
):

    transcript_repair_classification = (
        "PILOT_TARGET_REPLACEMENT_WITHIN_EXISTING_CASE_SCOPE_IS_FEASIBLE"
    )

else:

    transcript_repair_classification = (
        "CASE_EVIDENCE_SCOPE_REPAIR_REQUIRED_BEFORE_PILOT_REPLACEMENT"
    )


# ============================================================
# 8) Repair Feasibility Registry
# ============================================================

REPAIR_FEASIBILITY_AUDIT_V1_DRAFT = {
    "version":
        "repair_feasibility_audit_v1_draft",

    "status":
        "INSPECTED_NO_MUTATION",

    "book_exercise": {
        "selected_case_count":
            len(
                exercise_evidence_scope_audit_rows
            ),

        "same_heading_fully_inside_existing_case_count":
            exercise_full_heading_already_inside_case_count,

        "case_scope_expansion_required_count":
            exercise_requires_case_scope_expansion_count,

        "missing_same_heading_record_total":
            exercise_missing_heading_records_total,

        "repair_classification":
            exercise_repair_classification,
    },

    "transcript_single_primary_failed_pilot": {
        "physical_file_document_count":
            len(
                failed_single_primary_file_document_ids
            ),

        "current_case_evidence_count":
            len(
                failed_single_primary_case_evidence_ids
            ),

        "full_file_inside_existing_case_evidence":
            transcript_file_fully_inside_case_evidence,

        "repair_classification":
            transcript_repair_classification,
    },

    "repair_candidate_integrity_failure_count":
        len(
            repair_candidate_integrity_failures
        ),

    "repair_execution_status":
        "NOT_STARTED",

    "question_generation_status":
        "NOT_STARTED",

    "expected_answer_generation_status":
        "NOT_STARTED",

    "artifact_write_status":
        "NONE",
}


# ============================================================
# 9) Validation gates
# ============================================================

exact_38_book_scope_audits_gate = (
    len(
        exercise_evidence_scope_audit_rows
    )
    == 38
)


book_same_structure_integrity_gate = (
    len(
        exercise_same_structure_failures
    )
    == 0
)


failed_book_case_present_gate = (
    failed_exercise_scope_audit[
        "golden_case_id"
    ]
    ==
    failed_exercise_case_id
)


transcript_failed_case_file_resolved_gate = (
    len(
        failed_single_primary_file_document_ids
    )
    > 0
)


all_transcript_top_alternatives_same_file_gate = all(
    row[
        "inside_same_physical_file"
    ]
    is True

    for row
    in failed_single_primary_alternative_scope_rows
)


repair_candidates_integrity_gate = (
    len(
        repair_candidate_integrity_failures
    )
    == 0
)


repair_classifications_resolved_gate = (
    exercise_repair_classification
    in {
        "CASE_EVIDENCE_SCOPE_REPAIR_REQUIRED_BEFORE_TARGET_REPAIR",
        "TARGET_METHOD_REPAIR_ONLY_IS_FEASIBLE",
    }

    and

    transcript_repair_classification
    in {
        "PILOT_TARGET_REPLACEMENT_WITHIN_EXISTING_CASE_SCOPE_IS_FEASIBLE",
        "CASE_EVIDENCE_SCOPE_REPAIR_REQUIRED_BEFORE_PILOT_REPLACEMENT",
    }
)


no_mutation_gate = (
    REPAIR_FEASIBILITY_AUDIT_V1_DRAFT[
        "repair_execution_status"
    ]
    ==
    "NOT_STARTED"
)


question_generation_not_started_gate = (
    REPAIR_FEASIBILITY_AUDIT_V1_DRAFT[
        "question_generation_status"
    ]
    ==
    "NOT_STARTED"
)


expected_answer_not_started_gate = (
    REPAIR_FEASIBILITY_AUDIT_V1_DRAFT[
        "expected_answer_generation_status"
    ]
    ==
    "NOT_STARTED"
)


no_artifact_write_gate = (
    REPAIR_FEASIBILITY_AUDIT_V1_DRAFT[
        "artifact_write_status"
    ]
    ==
    "NONE"
)


cell_33_gates = {
    "exact_38_book_exercise_evidence_scopes_audited":
        exact_38_book_scope_audits_gate,

    "all_proposed_book_records_preserve_work_edition_heading_path":
        book_same_structure_integrity_gate,

    "failed_book_pilot_present_in_scope_audit":
        failed_book_case_present_gate,

    "failed_transcript_pilot_physical_file_resolved":
        transcript_failed_case_file_resolved_gate,

    "all_inspected_transcript_alternatives_remain_in_same_physical_file":
        all_transcript_top_alternatives_same_file_gate,

    "all_potential_repair_records_have_locator_lineage_attribution_rights":
        repair_candidates_integrity_gate,

    "repair_feasibility_classifications_resolved":
        repair_classifications_resolved_gate,

    "repair_execution_not_started":
        no_mutation_gate,

    "question_generation_not_started":
        question_generation_not_started_gate,

    "expected_answer_generation_not_started":
        expected_answer_not_started_gate,

    "no_artifact_write":
        no_artifact_write_gate,
}


cell_33_pass = all(
    cell_33_gates.values()
)


# ============================================================
# 10) Output
# ============================================================

print(
    "=== REPAIR FEASIBILITY & EVIDENCE-SCOPE AUDIT ==="
)


print(
    "\n=== BOOK / EXERCISE_OR_SELF_ASSESSMENT ==="
)

print(
    "Cases audited:",
    len(
        exercise_evidence_scope_audit_rows
    )
)

print(
    "Same-heading group already fully inside current case evidence:",
    exercise_full_heading_already_inside_case_count
)

print(
    "Cases requiring case-evidence expansion:",
    exercise_requires_case_scope_expansion_count
)

print(
    "Total same-heading records currently outside case evidence:",
    exercise_missing_heading_records_total
)

print(
    "Repair classification:",
    exercise_repair_classification
)


print(
    "\nFailed Pilot evidence-scope detail:"
)

print(
    json.dumps(
        failed_exercise_scope_audit,
        ensure_ascii=False,
        sort_keys=True,
    )
)


print(
    "\n=== TRANSCRIPT / SINGLE_SPEAKER_PRIMARY FAILED PILOT ==="
)

print(
    "Physical-file records:",
    len(
        failed_single_primary_file_document_ids
    )
)

print(
    "Current Golden Case evidence records:",
    len(
        failed_single_primary_case_evidence_ids
    )
)

print(
    "Full physical file inside current case evidence:",
    transcript_file_fully_inside_case_evidence
)

print(
    "Missing file records from case evidence:",
    len(
        transcript_file_missing_from_case_evidence
    )
)

print(
    "Case evidence outside physical file:",
    len(
        transcript_case_extra_outside_file
    )
)

print(
    "Repair classification:",
    transcript_repair_classification
)


print(
    "\nTop alternative scope eligibility:"
)

for row in (
    failed_single_primary_alternative_scope_rows
):

    print(
        "- rank:",
        row[
            "rank"
        ],
        "| doc:",
        row[
            "document_id"
        ],
        "| in_case:",
        row[
            "inside_current_golden_case_evidence"
        ],
        "| same_file:",
        row[
            "inside_same_physical_file"
        ],
        "| tokens:",
        row[
            "token_count"
        ],
        "| preview:",
        row[
            "text_preview"
        ],
    )


print(
    "\n=== REPAIR RECORD INTEGRITY ==="
)

print(
    "Integrity failures:",
    len(
        repair_candidate_integrity_failures
    )
)


if (
    repair_candidate_integrity_failures
):

    for failure in (
        repair_candidate_integrity_failures[
            :20
        ]
    ):

        print(
            "-",
            failure
        )


print(
    "\n=== CODE CELL 33 GATES ==="
)

for name, passed in (
    cell_33_gates.items()
):

    print(
        name,
        ":",
        "PASS"
        if passed
        else "FAIL"
    )


print(
    "\nCODE CELL 33 GATE:",
    "PASS"
    if cell_33_pass
    else "FAIL"
)


assert cell_33_pass, (
    "Repair feasibility audit failed. "
    "Do not mutate Golden Case evidence or Target Evidence until "
    "the permitted repair scope is explicitly established."
)
```

<div style="direction: rtl; white-space: normal; line-height: 1.7;">

<h3>Exercise Stratum Structural Identity and Boundary Discovery</h3>

<p>
Cell 33 ثابت کرد که Target تک‌رکوردی فعلی برای
<code>exercise_or_self_assessment</code>
کافی نیست و هر ۳۸ Case به Evidence بیشتری در همان Heading Path دسترسی دارند.
اما هنوز نباید تمام Heading Group به Case Evidence اضافه شود.
</p>

<p>
دلیل مهم این است که Heading Path ممکن است علاوه بر Exercise Core،
رکوردهای دیگری مانند منابع آموزشی، نام نرم‌افزار، URL یا توضیح ابزار را نیز در خود نگه دارد.
بنابراین <strong>Heading Path ≠ Exercise Boundary</strong>.
</p>

<p>
پیش از Repair باید یک مسئله‌ی بنیادی دیگر نیز بررسی شود:
آیا ۳۸ Golden Case انتخاب‌شده واقعاً ۳۸ Exercise مستقل هستند،
یا چند Case مختلف فقط رکوردهای متفاوتی از یک Exercise/Worksheet واحد هستند؟
</p>

<p>
این Cell سه چیز را فقط Discover می‌کند:
</p>

<ul>
<li>تعداد Exercise Heading Groupهای یکتا در میان ۳۸ Case؛</li>
<li>تعداد Caseهایی که یک Heading Group مشترک دارند؛</li>
<li>ساختار داخلی هر Group شامل Heading، Narrative، Table و URL-like runs.</li>
</ul>

<p>
همچنین جای Candidate فعلی داخل Group و Cueهای صریح مانند
<code>تمرین</code>، <code>کاربرگ</code>، <code>خودارزیابی</code> و <code>گام</code>
فقط به‌صورت توصیفی گزارش می‌شوند.
</p>

<p>
در این مرحله هیچ Exercise Boundary ساخته نمی‌شود،
هیچ Golden Case ادغام یا حذف نمی‌شود،
هیچ Evidence Scope یا Target تغییر نمی‌کند،
و Question Generation همچنان شروع نمی‌شود.
</p>

</div>

```python
# Code Cell 34

# ============================================================
# Exercise Stratum Structural Identity & Boundary Discovery
#
# DISCOVERY / INSPECTION ONLY.
#
# NO Golden Case mutation.
# NO deduplication.
# NO evidence expansion.
# NO target repair.
# NO pilot replacement.
# NO question generation.
# NO expected-answer generation.
# NO artifact writes.
# ============================================================


# ============================================================
# 1) Transparent descriptive signals
#
# These are NOT acceptance or boundary rules.
# They are used only to expose observable structure.
# ============================================================

EXERCISE_DESCRIPTIVE_CUES = [
    "تمرین",
    "کاربرگ",
    "خودارزیابی",
    "خود ارزیابی",
    "خودآزمایی",
    "خود آزمایی",
    "پرسشنامه",
    "گام",
]


URL_LIKE_PATTERN = re.compile(
    r"(?i)(?:https?://|www\.|[A-Za-z0-9.-]+\.(?:ir|com|org|net)(?:/|\b))"
)


def descriptive_cue_hits(
    text,
):

    normalized = str(
        text
        or ""
    )

    return [
        cue
        for cue
        in EXERCISE_DESCRIPTIVE_CUES
        if (
            cue
            in normalized
        )
    ]


def is_url_like_text(
    text,
):

    return bool(
        URL_LIKE_PATTERN.search(
            str(
                text
                or ""
            )
        )
    )


# ============================================================
# 2) Resolve each selected exercise Case to an exact
# Work / Edition / Heading-Path structural group
# ============================================================

exercise_selected_structure_rows = []


for case_record in (
    exercise_selected_cases
):

    case_id = (
        case_record[
            "golden_case_id"
        ]
    )

    candidate = (
        case_record[
            "candidate_snapshot"
        ]
    )

    candidate_document_id = (
        candidate[
            "document_id"
        ]
    )

    source_record = (
        corpus_target_index[
            candidate_document_id
        ]
    )

    locator = (
        source_record[
            "locator"
        ]
        or {}
    )

    metadata = (
        source_record[
            "source_metadata"
        ]
        or {}
    )


    heading_path = (
        locator.get(
            "heading_path"
        )
    )

    if (
        heading_path
        is None
    ):

        heading_path = (
            metadata.get(
                "heading_path"
            )
        )


    if isinstance(
        heading_path,
        list,
    ):

        heading_path_key = tuple(
            str(
                value
            )
            for value
            in heading_path
        )

    elif (
        heading_path
        is None
    ):

        heading_path_key = tuple()

    else:

        heading_path_key = (
            str(
                heading_path
            ),
        )


    structure_group_key = (
        locator.get(
            "work_id"
        ),
        locator.get(
            "edition_id"
        ),
        heading_path_key,
    )


    group_records = (
        book_heading_groups[
            structure_group_key
        ]
    )


    candidate_position = next(
        index
        for index, record
        in enumerate(
            group_records
        )
        if (
            record[
                "document_id"
            ]
            ==
            candidate_document_id
        )
    )


    exercise_selected_structure_rows.append(
        {
            "golden_case_id":
                case_id,

            "candidate_document_id":
                candidate_document_id,

            "work_id":
                structure_group_key[
                    0
                ],

            "edition_id":
                structure_group_key[
                    1
                ],

            "heading_path":
                list(
                    heading_path_key
                ),

            "structure_group_key":
                structure_group_key,

            "candidate_position":
                candidate_position,

            "candidate_paragraph_number":
                group_records[
                    candidate_position
                ][
                    "paragraph_number"
                ],

            "candidate_tokens":
                group_records[
                    candidate_position
                ][
                    "token_count"
                ],

            "candidate_in_table":
                group_records[
                    candidate_position
                ][
                    "in_table"
                ],

            "candidate_is_heading":
                group_records[
                    candidate_position
                ][
                    "is_heading"
                ],

            "candidate_url_like":
                is_url_like_text(
                    group_records[
                        candidate_position
                    ][
                        "text"
                    ]
                ),

            "candidate_cue_hits":
                descriptive_cue_hits(
                    group_records[
                        candidate_position
                    ][
                        "text"
                    ]
                ),

            "candidate_text":
                group_records[
                    candidate_position
                ][
                    "text"
                ],

            "group_record_count":
                len(
                    group_records
                ),
        }
    )


# ============================================================
# 3) Group selected Golden Cases by structural Exercise group
# ============================================================

selected_cases_by_exercise_group = defaultdict(
    list
)


for row in (
    exercise_selected_structure_rows
):

    selected_cases_by_exercise_group[
        row[
            "structure_group_key"
        ]
    ].append(
        row
    )


unique_selected_exercise_group_count = (
    len(
        selected_cases_by_exercise_group
    )
)


selected_case_count_per_group = Counter(
    len(
        rows
    )
    for rows
    in selected_cases_by_exercise_group.values()
)


groups_with_multiple_selected_cases = {
    group_key:
        rows

    for group_key, rows
    in selected_cases_by_exercise_group.items()

    if (
        len(
            rows
        )
        > 1
    )
}


selected_cases_inside_reused_groups = sum(
    len(
        rows
    )
    for rows
    in groups_with_multiple_selected_cases.values()
)


# ============================================================
# 4) Build structural run representation for each unique group
#
# Run signature consists only of observable metadata:
# - HEADING
# - TABLE
# - URL_LIKE
# - NARRATIVE
#
# This is descriptive only.
# ============================================================

def structural_record_class(
    record,
):

    if (
        record[
            "is_heading"
        ]
        is True
    ):

        return "HEADING"

    if (
        record[
            "in_table"
        ]
        is True
    ):

        return "TABLE"

    if is_url_like_text(
        record[
            "text"
        ]
    ):

        return "URL_LIKE"

    return "NARRATIVE"


def build_structural_runs(
    records,
):

    runs = []

    current_run = None


    for index, record in enumerate(
        records
    ):

        record_class = (
            structural_record_class(
                record
            )
        )


        if (
            current_run
            is None

            or

            current_run[
                "record_class"
            ]
            != record_class
        ):

            if (
                current_run
                is not None
            ):

                runs.append(
                    current_run
                )


            current_run = {
                "record_class":
                    record_class,

                "start_index":
                    index,

                "end_index":
                    index,

                "start_paragraph":
                    record[
                        "paragraph_number"
                    ],

                "end_paragraph":
                    record[
                        "paragraph_number"
                    ],

                "record_count":
                    1,

                "token_count":
                    record[
                        "token_count"
                    ],

                "document_ids": [
                    record[
                        "document_id"
                    ]
                ],
            }

        else:

            current_run[
                "end_index"
            ] = index

            current_run[
                "end_paragraph"
            ] = (
                record[
                    "paragraph_number"
                ]
            )

            current_run[
                "record_count"
            ] += 1

            current_run[
                "token_count"
            ] += (
                record[
                    "token_count"
                ]
            )

            current_run[
                "document_ids"
            ].append(
                record[
                    "document_id"
                ]
            )


    if (
        current_run
        is not None
    ):

        runs.append(
            current_run
        )


    return runs


# ============================================================
# 5) Unique group audit
# ============================================================

exercise_group_boundary_discovery_rows = []


for group_key, selected_rows in sorted(
    selected_cases_by_exercise_group.items(),
    key=lambda item: (
        str(
            item[
                0
            ][
                0
            ]
        ),
        str(
            item[
                0
            ][
                1
            ]
        ),
        item[
            0
        ][
            2
        ],
    ),
):

    group_records = (
        book_heading_groups[
            group_key
        ]
    )


    structural_runs = (
        build_structural_runs(
            group_records
        )
    )


    heading_record_count = sum(
        1
        for record
        in group_records
        if (
            record[
                "is_heading"
            ]
            is True
        )
    )


    table_record_count = sum(
        1
        for record
        in group_records
        if (
            record[
                "in_table"
            ]
            is True
        )
    )


    url_like_record_count = sum(
        1
        for record
        in group_records
        if is_url_like_text(
            record[
                "text"
            ]
        )
    )


    cue_records = []


    for index, record in enumerate(
        group_records
    ):

        hits = (
            descriptive_cue_hits(
                record[
                    "text"
                ]
            )
        )

        if hits:

            cue_records.append(
                {
                    "index":
                        index,

                    "paragraph_number":
                        record[
                            "paragraph_number"
                        ],

                    "document_id":
                        record[
                            "document_id"
                        ],

                    "cue_hits":
                        hits,

                    "text_preview":
                        record[
                            "text"
                        ][
                            :250
                        ],
                }
            )


    selected_positions = sorted(
        row[
            "candidate_position"
        ]
        for row
        in selected_rows
    )


    selected_document_ids = [
        row[
            "candidate_document_id"
        ]
        for row
        in sorted(
            selected_rows,
            key=lambda row:
                row[
                    "candidate_position"
                ],
        )
    ]


    sequence = []


    for index, record in enumerate(
        group_records
    ):

        sequence.append(
            {
                "index":
                    index,

                "paragraph_number":
                    record[
                        "paragraph_number"
                    ],

                "document_id":
                    record[
                        "document_id"
                    ],

                "record_class":
                    structural_record_class(
                        record
                    ),

                "token_count":
                    record[
                        "token_count"
                    ],

                "selected_as_golden_candidate":
                    (
                        record[
                            "document_id"
                        ]
                        in
                        set(
                            selected_document_ids
                        )
                    ),

                "cue_hits":
                    descriptive_cue_hits(
                        record[
                            "text"
                        ]
                    ),

                "text_preview":
                    record[
                        "text"
                    ][
                        :220
                    ],
            }
        )


    exercise_group_boundary_discovery_rows.append(
        {
            "work_id":
                group_key[
                    0
                ],

            "edition_id":
                group_key[
                    1
                ],

            "heading_path":
                list(
                    group_key[
                        2
                    ]
                ),

            "structure_group_key":
                group_key,

            "selected_golden_case_count":
                len(
                    selected_rows
                ),

            "selected_golden_case_ids":
                [
                    row[
                        "golden_case_id"
                    ]
                    for row
                    in selected_rows
                ],

            "selected_candidate_positions":
                selected_positions,

            "selected_candidate_document_ids":
                selected_document_ids,

            "group_record_count":
                len(
                    group_records
                ),

            "group_total_tokens":
                sum(
                    record[
                        "token_count"
                    ]
                    for record
                    in group_records
                ),

            "heading_record_count":
                heading_record_count,

            "table_record_count":
                table_record_count,

            "url_like_record_count":
                url_like_record_count,

            "cue_record_count":
                len(
                    cue_records
                ),

            "cue_records":
                cue_records,

            "structural_runs":
                structural_runs,

            "record_sequence":
                sequence,

            "exercise_boundary_status":
                "NOT_YET_DEFINED",
        }
    )


# ============================================================
# 6) Structural pattern distributions
# ============================================================

group_record_count_values = [
    row[
        "group_record_count"
    ]
    for row
    in exercise_group_boundary_discovery_rows
]


group_token_values = [
    row[
        "group_total_tokens"
    ]
    for row
    in exercise_group_boundary_discovery_rows
]


group_table_record_values = [
    row[
        "table_record_count"
    ]
    for row
    in exercise_group_boundary_discovery_rows
]


group_url_like_values = [
    row[
        "url_like_record_count"
    ]
    for row
    in exercise_group_boundary_discovery_rows
]


structural_run_signature_counts = Counter(
    tuple(
        run[
            "record_class"
        ]
        for run
        in row[
            "structural_runs"
        ]
    )
    for row
    in exercise_group_boundary_discovery_rows
)


# ============================================================
# 7) Selected-record semantic duplication risk audit
#
# This does NOT call them duplicates.
# It only exposes whether multiple Golden Cases refer to the
# same Work / Edition / Heading Path.
# ============================================================

same_group_multi_case_rows = []


for row in (
    exercise_group_boundary_discovery_rows
):

    if (
        row[
            "selected_golden_case_count"
        ]
        <= 1
    ):
        continue


    same_group_multi_case_rows.append(
        {
            "work_id":
                row[
                    "work_id"
                ],

            "edition_id":
                row[
                    "edition_id"
                ],

            "heading_path":
                row[
                    "heading_path"
                ],

            "selected_golden_case_count":
                row[
                    "selected_golden_case_count"
                ],

            "selected_candidate_positions":
                row[
                    "selected_candidate_positions"
                ],

            "selected_candidate_document_ids":
                row[
                    "selected_candidate_document_ids"
                ],
        }
    )


# ============================================================
# 8) Inspect failed Pilot's unique group in full
# ============================================================

failed_exercise_structure_row = next(
    row
    for row
    in exercise_group_boundary_discovery_rows
    if (
        failed_exercise_case_id
        in
        row[
            "selected_golden_case_ids"
        ]
    )
)


# ============================================================
# 9) Discovery registry
# ============================================================

EXERCISE_BOUNDARY_DISCOVERY_V1_DRAFT = {
    "version":
        "exercise_boundary_discovery_v1_draft",

    "status":
        "STRUCTURAL_IDENTITY_AND_BOUNDARY_SIGNALS_DISCOVERED",

    "selected_case_count":
        len(
            exercise_selected_structure_rows
        ),

    "unique_structural_group_count":
        unique_selected_exercise_group_count,

    "groups_with_multiple_selected_cases":
        len(
            groups_with_multiple_selected_cases
        ),

    "selected_cases_inside_reused_groups":
        selected_cases_inside_reused_groups,

    "boundary_definition_status":
        "NOT_YET_DEFINED",

    "deduplication_status":
        "NOT_STARTED",

    "case_identity_repair_status":
        "NOT_STARTED",

    "evidence_scope_repair_status":
        "NOT_STARTED",

    "target_repair_status":
        "NOT_STARTED",

    "question_generation_status":
        "NOT_STARTED",

    "expected_answer_generation_status":
        "NOT_STARTED",

    "artifact_write_status":
        "NONE",
}


# ============================================================
# 10) Validation gates
# ============================================================

exact_38_selected_cases_resolved_gate = (
    len(
        exercise_selected_structure_rows
    )
    == 38
)


all_cases_resolve_valid_group_gate = all(
    row[
        "group_record_count"
    ]
    >= 1

    and

    row[
        "candidate_position"
    ]
    >= 0

    for row
    in exercise_selected_structure_rows
)


all_selected_cases_accounted_in_groups_gate = (
    sum(
        row[
            "selected_golden_case_count"
        ]
        for row
        in exercise_group_boundary_discovery_rows
    )
    == 38
)


all_unique_groups_have_structural_runs_gate = all(
    len(
        row[
            "structural_runs"
        ]
    )
    >= 1

    for row
    in exercise_group_boundary_discovery_rows
)


all_unique_groups_have_record_sequences_gate = all(
    len(
        row[
            "record_sequence"
        ]
    )
    ==
    row[
        "group_record_count"
    ]

    for row
    in exercise_group_boundary_discovery_rows
)


failed_pilot_group_resolved_gate = (
    failed_exercise_case_id
    in
    failed_exercise_structure_row[
        "selected_golden_case_ids"
    ]
)


boundary_not_prejudged_gate = all(
    row[
        "exercise_boundary_status"
    ]
    ==
    "NOT_YET_DEFINED"

    for row
    in exercise_group_boundary_discovery_rows
)


no_case_identity_mutation_gate = (
    EXERCISE_BOUNDARY_DISCOVERY_V1_DRAFT[
        "case_identity_repair_status"
    ]
    ==
    "NOT_STARTED"
)


no_evidence_scope_mutation_gate = (
    EXERCISE_BOUNDARY_DISCOVERY_V1_DRAFT[
        "evidence_scope_repair_status"
    ]
    ==
    "NOT_STARTED"
)


no_target_mutation_gate = (
    EXERCISE_BOUNDARY_DISCOVERY_V1_DRAFT[
        "target_repair_status"
    ]
    ==
    "NOT_STARTED"
)


question_generation_not_started_gate = (
    EXERCISE_BOUNDARY_DISCOVERY_V1_DRAFT[
        "question_generation_status"
    ]
    ==
    "NOT_STARTED"
)


expected_answer_not_started_gate = (
    EXERCISE_BOUNDARY_DISCOVERY_V1_DRAFT[
        "expected_answer_generation_status"
    ]
    ==
    "NOT_STARTED"
)


no_artifact_write_gate = (
    EXERCISE_BOUNDARY_DISCOVERY_V1_DRAFT[
        "artifact_write_status"
    ]
    ==
    "NONE"
)


cell_34_gates = {
    "exact_38_selected_exercise_cases_resolved":
        exact_38_selected_cases_resolved_gate,

    "all_selected_cases_resolve_valid_structural_group":
        all_cases_resolve_valid_group_gate,

    "all_38_selected_cases_accounted_for_in_unique_groups":
        all_selected_cases_accounted_in_groups_gate,

    "all_unique_groups_have_structural_runs":
        all_unique_groups_have_structural_runs_gate,

    "all_unique_groups_have_complete_record_sequence":
        all_unique_groups_have_record_sequences_gate,

    "failed_pilot_structural_group_resolved":
        failed_pilot_group_resolved_gate,

    "exercise_boundary_not_prejudged":
        boundary_not_prejudged_gate,

    "case_identity_repair_not_started":
        no_case_identity_mutation_gate,

    "evidence_scope_repair_not_started":
        no_evidence_scope_mutation_gate,

    "target_repair_not_started":
        no_target_mutation_gate,

    "question_generation_not_started":
        question_generation_not_started_gate,

    "expected_answer_generation_not_started":
        expected_answer_not_started_gate,

    "no_artifact_write":
        no_artifact_write_gate,
}


cell_34_pass = all(
    cell_34_gates.values()
)


# ============================================================
# 11) Output
# ============================================================

print(
    "=== EXERCISE STRUCTURAL IDENTITY & BOUNDARY DISCOVERY ==="
)


print(
    "Selected Golden Cases:",
    len(
        exercise_selected_structure_rows
    )
)

print(
    "Unique Work/Edition/Heading groups:",
    unique_selected_exercise_group_count
)

print(
    "Groups containing multiple selected Golden Cases:",
    len(
        groups_with_multiple_selected_cases
    )
)

print(
    "Selected cases inside reused groups:",
    selected_cases_inside_reused_groups
)


print(
    "\n=== SELECTED CASES PER STRUCTURAL GROUP ==="
)

for case_count, group_count in sorted(
    selected_case_count_per_group.items()
):

    print(
        "- selected cases per group:",
        case_count,
        "| groups:",
        group_count,
    )


print(
    "\n=== UNIQUE GROUP SIZE DISTRIBUTIONS ==="
)

print(
    "Record count:",
    {
        "min":
            min(
                group_record_count_values
            ),

        "median":
            empirical_nearest_rank(
                group_record_count_values,
                0.50,
            ),

        "max":
            max(
                group_record_count_values
            ),
    }
)

print(
    "Token count:",
    {
        "min":
            min(
                group_token_values
            ),

        "median":
            empirical_nearest_rank(
                group_token_values,
                0.50,
            ),

        "max":
            max(
                group_token_values
            ),
    }
)

print(
    "Table records:",
    {
        "min":
            min(
                group_table_record_values
            ),

        "median":
            empirical_nearest_rank(
                group_table_record_values,
                0.50,
            ),

        "max":
            max(
                group_table_record_values
            ),
    }
)

print(
    "URL-like records:",
    {
        "min":
            min(
                group_url_like_values
            ),

        "median":
            empirical_nearest_rank(
                group_url_like_values,
                0.50,
            ),

        "max":
            max(
                group_url_like_values
            ),
    }
)


print(
    "\n=== STRUCTURAL RUN SIGNATURES ==="
)

for signature, count in (
    structural_run_signature_counts.most_common()
):

    print(
        "-",
        signature,
        ":",
        count,
    )


print(
    "\n=== GROUPS WITH MULTIPLE SELECTED GOLDEN CASES ==="
)

if (
    same_group_multi_case_rows
):

    for row in (
        same_group_multi_case_rows
    ):

        print(
            json.dumps(
                row,
                ensure_ascii=False,
                sort_keys=True,
            )
        )

else:

    print(
        "NONE"
    )


print(
    "\n=== FAILED PILOT FULL STRUCTURAL SEQUENCE ==="
)

print(
    "Heading path:",
    failed_exercise_structure_row[
        "heading_path"
    ]
)

print(
    "Selected Golden Cases in this group:",
    failed_exercise_structure_row[
        "selected_golden_case_count"
    ]
)

print(
    "Selected candidate positions:",
    failed_exercise_structure_row[
        "selected_candidate_positions"
    ]
)

print(
    "Structural runs:"
)

for run in (
    failed_exercise_structure_row[
        "structural_runs"
    ]
):

    print(
        json.dumps(
            {
                "record_class":
                    run[
                        "record_class"
                    ],

                "start_index":
                    run[
                        "start_index"
                    ],

                "end_index":
                    run[
                        "end_index"
                    ],

                "start_paragraph":
                    run[
                        "start_paragraph"
                    ],

                "end_paragraph":
                    run[
                        "end_paragraph"
                    ],

                "record_count":
                    run[
                        "record_count"
                    ],

                "token_count":
                    run[
                        "token_count"
                    ],
            },
            ensure_ascii=False,
            sort_keys=True,
        )
    )


print(
    "\nRecord sequence:"
)

for item in (
    failed_exercise_structure_row[
        "record_sequence"
    ]
):

    print(
        "- index:",
        item[
            "index"
        ],
        "| paragraph:",
        item[
            "paragraph_number"
        ],
        "| class:",
        item[
            "record_class"
        ],
        "| selected:",
        item[
            "selected_as_golden_candidate"
        ],
        "| cues:",
        item[
            "cue_hits"
        ],
        "| tokens:",
        item[
            "token_count"
        ],
        "| text:",
        item[
            "text_preview"
        ],
    )


print(
    "\n=== ALL UNIQUE GROUP SUMMARIES ==="
)

for index, row in enumerate(
    exercise_group_boundary_discovery_rows,
    start=1,
):

    print(
        "\nGROUP",
        index,
        "/",
        len(
            exercise_group_boundary_discovery_rows
        )
    )

    print(
        "Heading:",
        row[
            "heading_path"
        ]
    )

    print(
        "Selected cases:",
        row[
            "selected_golden_case_count"
        ]
    )

    print(
        "Selected positions:",
        row[
            "selected_candidate_positions"
        ]
    )

    print(
        "Records:",
        row[
            "group_record_count"
        ],
        "| tokens:",
        row[
            "group_total_tokens"
        ],
        "| table records:",
        row[
            "table_record_count"
        ],
        "| URL-like:",
        row[
            "url_like_record_count"
        ],
        "| cue records:",
        row[
            "cue_record_count"
        ],
    )

    print(
        "Run signature:",
        tuple(
            run[
                "record_class"
            ]
            for run
            in row[
                "structural_runs"
            ]
        )
    )

    print(
        "Cue records:"
    )

    for cue_record in (
        row[
            "cue_records"
        ]
    ):

        print(
            "  -",
            cue_record[
                "paragraph_number"
            ],
            "|",
            cue_record[
                "cue_hits"
            ],
            "|",
            cue_record[
                "text_preview"
            ],
        )


print(
    "\n=== CODE CELL 34 GATES ==="
)

for name, passed in (
    cell_34_gates.items()
):

    print(
        name,
        ":",
        "PASS"
        if passed
        else "FAIL"
    )


print(
    "\nCODE CELL 34 GATE:",
    "PASS"
    if cell_34_pass
    else "FAIL"
)


assert cell_34_pass, (
    "Exercise structural identity/boundary discovery failed. "
    "Do not expand Golden Case evidence or repair exercise targets "
    "until exercise-level identity and structural group reuse are "
    "fully observable."
)
```

<div style="direction: rtl; white-space: normal; line-height: 1.7;">

<h3>Exercise Unit Boundary Candidate Discovery</h3>

<p>
Cell 34 نشان داد ۳۸ Case رکوردی تنها به ۱۸ واحد ساختاری مستقل
<code>Work + Edition + Heading Path</code>
برمی‌گردند.
بنابراین پیش از Repair باید مرز واقعی هر Exercise/Worksheet در داخل Heading Group مشخص شود.
</p>

<p>
Heading Group به‌تنهایی Boundary کافی نیست، زیرا پس از بخش اصلی تمرین ممکن است
بخش‌هایی مانند «مطالعه بیشتر»، «منابع بیشتر»، URL، معرفی سامانه یا محتوای مکمل قرار گرفته باشد.
</p>

<p>
در این مرحله فقط <strong>Boundary Candidate</strong> کشف می‌شود.
هیچ Boundary نهایی Freeze نمی‌شود.
</p>

<h4>منطق Discovery</h4>

<ul>
<li>شروع واحد Exercise همان Heading اصلی Group است.</li>
<li>وجود Table یا Narrative بعد از Heading به‌خودی‌خود پایان Exercise نیست.</li>
<li>عبارت‌های صریح انتقالی مانند «مطالعه بیشتر» و «منابع بیشتر» Candidate قوی برای شروع Supplemental Material هستند.</li>
<li>URL به‌تنهایی Boundary قطعی نیست.</li>
<li>اگر Cue صریح وجود نداشته باشد، هیچ Boundary حدسی ساخته نمی‌شود.</li>
<li>Candidate Recordهای انتخاب‌شده‌ی قبلی برای تعیین Boundary به‌عنوان Ground Truth استفاده نمی‌شوند.</li>
</ul>

<p>
هدف Cell تعیین این است که برای چند مورد از ۱۸ Exercise Group می‌توان
یک Boundary صریح و بازتولیدپذیر پیدا کرد و برای چند Group هنوز Human/Structural Review لازم است.
</p>

<p>
هیچ Golden Case، Evidence Scope، Target، Question یا Expected Answer تغییر نمی‌کند.
</p>

</div>

```python
# Code Cell 35

# ============================================================
# Exercise Unit Boundary Candidate Discovery
#
# DISCOVERY ONLY.
#
# NO final boundary freeze.
# NO Golden Case merge/delete.
# NO evidence-scope mutation.
# NO target mutation.
# NO question generation.
# NO expected-answer generation.
# NO artifact writes.
# ============================================================


# ============================================================
# 1) Explicit transition markers
#
# These markers are intentionally narrow.
# They indicate a possible transition from the exercise core
# to supplemental material.
#
# URL presence alone is NOT treated as a boundary.
# ============================================================

EXPLICIT_SUPPLEMENTAL_BOUNDARY_MARKERS = [
    "مطالعه بیشتر",
    "مطالعۀ بیشتر",
    "برای مطالعه بیشتر",
    "برای مطالعۀ بیشتر",
    "منابع بیشتر",
    "منبع بیشتر",
]


def normalized_boundary_text(
    text,
):

    return (
        str(
            text
            or ""
        )
        .replace(
            "\u200c",
            " "
        )
        .strip()
    )


def explicit_supplemental_marker_hits(
    text,
):

    normalized = (
        normalized_boundary_text(
            text
        )
    )

    return [
        marker
        for marker
        in EXPLICIT_SUPPLEMENTAL_BOUNDARY_MARKERS
        if (
            normalized
            ==
            normalized_boundary_text(
                marker
            )

            or

            normalized.startswith(
                normalized_boundary_text(
                    marker
                )
            )
        )
    ]


# ============================================================
# 2) Discover explicit boundary candidate per unique group
# ============================================================

exercise_boundary_candidate_rows = []


for group_row in (
    exercise_group_boundary_discovery_rows
):

    group_records = (
        book_heading_groups[
            group_row[
                "structure_group_key"
            ]
        ]
    )


    explicit_candidates = []


    for index, record in enumerate(
        group_records
    ):

        marker_hits = (
            explicit_supplemental_marker_hits(
                record[
                    "text"
                ]
            )
        )

        if not marker_hits:
            continue


        explicit_candidates.append(
            {
                "index":
                    index,

                "paragraph_number":
                    record[
                        "paragraph_number"
                    ],

                "document_id":
                    record[
                        "document_id"
                    ],

                "marker_hits":
                    marker_hits,

                "record_class":
                    structural_record_class(
                        record
                    ),

                "text":
                    record[
                        "text"
                    ],
            }
        )


    # --------------------------------------------------------
    # Candidate rule:
    # earliest explicit supplemental marker after heading.
    #
    # This is still only a candidate boundary.
    # --------------------------------------------------------

    if explicit_candidates:

        selected_candidate = sorted(
            explicit_candidates,
            key=lambda item: (
                item[
                    "index"
                ],
                item[
                    "document_id"
                ],
            ),
        )[0]


        supplemental_start_index = (
            selected_candidate[
                "index"
            ]
        )


        exercise_core_end_index = (
            supplemental_start_index
            - 1
        )


        if (
            exercise_core_end_index
            < 0
        ):

            boundary_status = (
                "INVALID_EXPLICIT_BOUNDARY_CANDIDATE"
            )

        else:

            boundary_status = (
                "EXPLICIT_BOUNDARY_CANDIDATE_DISCOVERED"
            )


    else:

        selected_candidate = None

        supplemental_start_index = None

        exercise_core_end_index = None

        boundary_status = (
            "NO_EXPLICIT_BOUNDARY_CANDIDATE"
        )


    # --------------------------------------------------------
    # Build candidate exercise-core preview.
    # No mutation.
    # --------------------------------------------------------

    if (
        exercise_core_end_index
        is not None

        and

        exercise_core_end_index
        >= 0
    ):

        candidate_core_records = (
            group_records[
                :exercise_core_end_index + 1
            ]
        )

    else:

        candidate_core_records = []


    candidate_core_document_ids = [
        record[
            "document_id"
        ]
        for record
        in candidate_core_records
    ]


    candidate_core_tokens = sum(
        record[
            "token_count"
        ]
        for record
        in candidate_core_records
    )


    selected_case_positions_inside_candidate_core = [
        position
        for position
        in group_row[
            "selected_candidate_positions"
        ]
        if (
            exercise_core_end_index
            is not None

            and

            position
            <= exercise_core_end_index
        )
    ]


    selected_case_positions_after_candidate_boundary = [
        position
        for position
        in group_row[
            "selected_candidate_positions"
        ]
        if (
            supplemental_start_index
            is not None

            and

            position
            >= supplemental_start_index
        )
    ]


    exercise_boundary_candidate_rows.append(
        {
            "work_id":
                group_row[
                    "work_id"
                ],

            "edition_id":
                group_row[
                    "edition_id"
                ],

            "heading_path":
                group_row[
                    "heading_path"
                ],

            "structure_group_key":
                group_row[
                    "structure_group_key"
                ],

            "group_record_count":
                group_row[
                    "group_record_count"
                ],

            "group_total_tokens":
                group_row[
                    "group_total_tokens"
                ],

            "selected_golden_case_count":
                group_row[
                    "selected_golden_case_count"
                ],

            "selected_candidate_positions":
                list(
                    group_row[
                        "selected_candidate_positions"
                    ]
                ),

            "explicit_boundary_candidates":
                explicit_candidates,

            "selected_explicit_boundary":
                selected_candidate,

            "exercise_core_start_index":
                0,

            "exercise_core_end_index":
                exercise_core_end_index,

            "supplemental_start_index":
                supplemental_start_index,

            "candidate_core_record_count":
                len(
                    candidate_core_records
                ),

            "candidate_core_token_count":
                candidate_core_tokens,

            "candidate_core_document_ids":
                candidate_core_document_ids,

            "selected_case_positions_inside_candidate_core":
                selected_case_positions_inside_candidate_core,

            "selected_case_positions_after_candidate_boundary":
                selected_case_positions_after_candidate_boundary,

            "boundary_status":
                boundary_status,

            "final_boundary_status":
                "NOT_FROZEN",
        }
    )


# ============================================================
# 3) Aggregate boundary discovery
# ============================================================

boundary_status_counts = Counter(
    row[
        "boundary_status"
    ]
    for row
    in exercise_boundary_candidate_rows
)


explicit_boundary_rows = [
    row
    for row
    in exercise_boundary_candidate_rows
    if (
        row[
            "boundary_status"
        ]
        ==
        "EXPLICIT_BOUNDARY_CANDIDATE_DISCOVERED"
    )
]


no_explicit_boundary_rows = [
    row
    for row
    in exercise_boundary_candidate_rows
    if (
        row[
            "boundary_status"
        ]
        ==
        "NO_EXPLICIT_BOUNDARY_CANDIDATE"
    )
]


invalid_boundary_rows = [
    row
    for row
    in exercise_boundary_candidate_rows
    if (
        row[
            "boundary_status"
        ]
        ==
        "INVALID_EXPLICIT_BOUNDARY_CANDIDATE"
    )
]


# ============================================================
# 4) Measure how many existing selected record-level Cases
# fall inside vs after an explicit candidate boundary.
#
# Descriptive only.
# This does NOT reject any Case yet.
# ============================================================

selected_positions_inside_explicit_core_total = sum(
    len(
        row[
            "selected_case_positions_inside_candidate_core"
        ]
    )
    for row
    in explicit_boundary_rows
)


selected_positions_after_explicit_boundary_total = sum(
    len(
        row[
            "selected_case_positions_after_candidate_boundary"
        ]
    )
    for row
    in explicit_boundary_rows
)


groups_with_selected_cases_after_boundary = [
    row
    for row
    in explicit_boundary_rows
    if (
        len(
            row[
                "selected_case_positions_after_candidate_boundary"
            ]
        )
        > 0
    )
]


# ============================================================
# 5) Failed Pilot boundary discovery
# ============================================================

failed_exercise_boundary_row = next(
    row
    for row
    in exercise_boundary_candidate_rows
    if (
        row[
            "structure_group_key"
        ]
        ==
        failed_exercise_structure_row[
            "structure_group_key"
        ]
    )
)


# ============================================================
# 6) Structural integrity of candidate cores
#
# Candidate core must:
# - contain the heading at index 0;
# - be contiguous from index 0;
# - remain inside same structural group.
# ============================================================

candidate_core_integrity_failures = []


for row in (
    explicit_boundary_rows
):

    group_records = (
        book_heading_groups[
            row[
                "structure_group_key"
            ]
        ]
    )


    expected_ids = [
        record[
            "document_id"
        ]
        for record
        in group_records[
            :row[
                "exercise_core_end_index"
            ] + 1
        ]
    ]


    if (
        row[
            "candidate_core_document_ids"
        ]
        !=
        expected_ids
    ):

        candidate_core_integrity_failures.append(
            {
                "heading_path":
                    row[
                        "heading_path"
                    ],

                "reason":
                    "NONCONTIGUOUS_CORE",
            }
        )


    if not (
        row[
            "candidate_core_document_ids"
        ]
    ):

        candidate_core_integrity_failures.append(
            {
                "heading_path":
                    row[
                        "heading_path"
                    ],

                "reason":
                    "EMPTY_CORE",
            }
        )

        continue


    first_record = (
        group_records[
            0
        ]
    )


    if (
        first_record[
            "is_heading"
        ]
        is not True
    ):

        candidate_core_integrity_failures.append(
            {
                "heading_path":
                    row[
                        "heading_path"
                    ],

                "reason":
                    "CORE_DOES_NOT_START_WITH_HEADING",
            }
        )


# ============================================================
# 7) Important non-decision:
#
# No-explicit-boundary groups remain unresolved.
# We do not fall back to URL, table-end, token count, or
# selected-record positions.
# ============================================================

unresolved_group_count = (
    len(
        no_explicit_boundary_rows
    )
)


# ============================================================
# 8) Boundary discovery registry
# ============================================================

EXERCISE_UNIT_BOUNDARY_CANDIDATE_REGISTRY_V1_DRAFT = {
    "version":
        "exercise_unit_boundary_candidate_registry_v1_draft",

    "status":
        "EXPLICIT_BOUNDARY_CANDIDATES_DISCOVERED_NOT_FROZEN",

    "unique_exercise_group_count":
        len(
            exercise_boundary_candidate_rows
        ),

    "explicit_boundary_candidate_group_count":
        len(
            explicit_boundary_rows
        ),

    "unresolved_group_count":
        unresolved_group_count,

    "invalid_boundary_candidate_group_count":
        len(
            invalid_boundary_rows
        ),

    "boundary_policy": (
        "EARLIEST_EXPLICIT_SUPPLEMENTAL_TRANSITION_MARKER_ONLY"
    ),

    "url_as_boundary":
        False,

    "table_end_as_boundary":
        False,

    "token_threshold_as_boundary":
        False,

    "selected_golden_record_as_boundary":
        False,

    "final_boundary_status":
        "NOT_FROZEN",

    "golden_case_identity_repair_status":
        "NOT_STARTED",

    "evidence_scope_repair_status":
        "NOT_STARTED",

    "target_repair_status":
        "NOT_STARTED",

    "question_generation_status":
        "NOT_STARTED",

    "expected_answer_generation_status":
        "NOT_STARTED",

    "artifact_write_status":
        "NONE",
}


# ============================================================
# 9) Validation gates
# ============================================================

exact_18_groups_audited_gate = (
    len(
        exercise_boundary_candidate_rows
    )
    == 18
)


all_groups_classified_gate = (
    sum(
        boundary_status_counts.values()
    )
    == 18
)


no_invalid_boundary_gate = (
    len(
        invalid_boundary_rows
    )
    == 0
)


explicit_candidate_core_integrity_gate = (
    len(
        candidate_core_integrity_failures
    )
    == 0
)


failed_pilot_boundary_resolved_gate = (
    failed_exercise_boundary_row[
        "boundary_status"
    ]
    in {
        "EXPLICIT_BOUNDARY_CANDIDATE_DISCOVERED",
        "NO_EXPLICIT_BOUNDARY_CANDIDATE",
    }
)


url_not_used_as_boundary_gate = (
    EXERCISE_UNIT_BOUNDARY_CANDIDATE_REGISTRY_V1_DRAFT[
        "url_as_boundary"
    ]
    is False
)


no_fallback_boundary_invention_gate = (
    EXERCISE_UNIT_BOUNDARY_CANDIDATE_REGISTRY_V1_DRAFT[
        "table_end_as_boundary"
    ]
    is False

    and

    EXERCISE_UNIT_BOUNDARY_CANDIDATE_REGISTRY_V1_DRAFT[
        "token_threshold_as_boundary"
    ]
    is False

    and

    EXERCISE_UNIT_BOUNDARY_CANDIDATE_REGISTRY_V1_DRAFT[
        "selected_golden_record_as_boundary"
    ]
    is False
)


boundary_not_frozen_gate = (
    EXERCISE_UNIT_BOUNDARY_CANDIDATE_REGISTRY_V1_DRAFT[
        "final_boundary_status"
    ]
    ==
    "NOT_FROZEN"

    and

    all(
        row[
            "final_boundary_status"
        ]
        ==
        "NOT_FROZEN"

        for row
        in exercise_boundary_candidate_rows
    )
)


no_case_identity_repair_gate = (
    EXERCISE_UNIT_BOUNDARY_CANDIDATE_REGISTRY_V1_DRAFT[
        "golden_case_identity_repair_status"
    ]
    ==
    "NOT_STARTED"
)


no_evidence_scope_repair_gate = (
    EXERCISE_UNIT_BOUNDARY_CANDIDATE_REGISTRY_V1_DRAFT[
        "evidence_scope_repair_status"
    ]
    ==
    "NOT_STARTED"
)


no_target_repair_gate = (
    EXERCISE_UNIT_BOUNDARY_CANDIDATE_REGISTRY_V1_DRAFT[
        "target_repair_status"
    ]
    ==
    "NOT_STARTED"
)


question_generation_not_started_gate = (
    EXERCISE_UNIT_BOUNDARY_CANDIDATE_REGISTRY_V1_DRAFT[
        "question_generation_status"
    ]
    ==
    "NOT_STARTED"
)


expected_answer_not_started_gate = (
    EXERCISE_UNIT_BOUNDARY_CANDIDATE_REGISTRY_V1_DRAFT[
        "expected_answer_generation_status"
    ]
    ==
    "NOT_STARTED"
)


no_artifact_write_gate = (
    EXERCISE_UNIT_BOUNDARY_CANDIDATE_REGISTRY_V1_DRAFT[
        "artifact_write_status"
    ]
    ==
    "NONE"
)


cell_35_gates = {
    "exact_18_unique_exercise_groups_audited":
        exact_18_groups_audited_gate,

    "all_18_groups_receive_boundary_discovery_status":
        all_groups_classified_gate,

    "no_invalid_explicit_boundary_candidate":
        no_invalid_boundary_gate,

    "all_explicit_candidate_cores_are_contiguous_and_structurally_valid":
        explicit_candidate_core_integrity_gate,

    "failed_pilot_group_boundary_discovery_resolved":
        failed_pilot_boundary_resolved_gate,

    "url_presence_not_used_as_boundary":
        url_not_used_as_boundary_gate,

    "no_table_token_or_selected_record_fallback_boundary_invented":
        no_fallback_boundary_invention_gate,

    "final_exercise_boundary_not_frozen":
        boundary_not_frozen_gate,

    "golden_case_identity_repair_not_started":
        no_case_identity_repair_gate,

    "evidence_scope_repair_not_started":
        no_evidence_scope_repair_gate,

    "target_repair_not_started":
        no_target_repair_gate,

    "question_generation_not_started":
        question_generation_not_started_gate,

    "expected_answer_generation_not_started":
        expected_answer_not_started_gate,

    "no_artifact_write":
        no_artifact_write_gate,
}


cell_35_pass = all(
    cell_35_gates.values()
)


# ============================================================
# 10) Output
# ============================================================

print(
    "=== EXERCISE UNIT BOUNDARY CANDIDATE DISCOVERY ==="
)

print(
    "Unique exercise groups:",
    len(
        exercise_boundary_candidate_rows
    )
)

print(
    "Explicit boundary candidate groups:",
    len(
        explicit_boundary_rows
    )
)

print(
    "Groups without explicit boundary candidate:",
    len(
        no_explicit_boundary_rows
    )
)

print(
    "Invalid explicit boundary candidates:",
    len(
        invalid_boundary_rows
    )
)


print(
    "\n=== BOUNDARY STATUS COUNTS ==="
)

for status, count in sorted(
    boundary_status_counts.items()
):

    print(
        "-",
        status,
        ":",
        count,
    )


print(
    "\n=== EXISTING SELECTED RECORD POSITIONS VS EXPLICIT BOUNDARY ==="
)

print(
    "Selected record-level cases inside candidate exercise core:",
    selected_positions_inside_explicit_core_total
)

print(
    "Selected record-level cases at/after candidate supplemental boundary:",
    selected_positions_after_explicit_boundary_total
)

print(
    "Groups with selected records after candidate boundary:",
    len(
        groups_with_selected_cases_after_boundary
    )
)


print(
    "\n=== EXPLICIT BOUNDARY CANDIDATES ==="
)

for row in (
    explicit_boundary_rows
):

    print(
        "\nHeading:",
        row[
            "heading_path"
        ]
    )

    print(
        "Selected Golden Cases:",
        row[
            "selected_golden_case_count"
        ]
    )

    print(
        "Boundary candidate:",
        json.dumps(
            row[
                "selected_explicit_boundary"
            ],
            ensure_ascii=False,
            sort_keys=True,
        )
    )

    print(
        "Candidate exercise core:",
        {
            "records":
                row[
                    "candidate_core_record_count"
                ],

            "tokens":
                row[
                    "candidate_core_token_count"
                ],

            "end_index":
                row[
                    "exercise_core_end_index"
                ],
        }
    )

    print(
        "Selected positions inside core:",
        row[
            "selected_case_positions_inside_candidate_core"
        ]
    )

    print(
        "Selected positions after boundary:",
        row[
            "selected_case_positions_after_candidate_boundary"
        ]
    )


print(
    "\n=== GROUPS WITHOUT EXPLICIT BOUNDARY CANDIDATE ==="
)

for row in (
    no_explicit_boundary_rows
):

    print(
        "-",
        row[
            "heading_path"
        ],
        "| records:",
        row[
            "group_record_count"
        ],
        "| tokens:",
        row[
            "group_total_tokens"
        ],
        "| selected cases:",
        row[
            "selected_golden_case_count"
        ],
    )


print(
    "\n=== FAILED PILOT BOUNDARY DISCOVERY ==="
)

print(
    json.dumps(
        {
            "heading_path":
                failed_exercise_boundary_row[
                    "heading_path"
                ],

            "boundary_status":
                failed_exercise_boundary_row[
                    "boundary_status"
                ],

            "selected_explicit_boundary":
                failed_exercise_boundary_row[
                    "selected_explicit_boundary"
                ],

            "exercise_core_end_index":
                failed_exercise_boundary_row[
                    "exercise_core_end_index"
                ],

            "candidate_core_record_count":
                failed_exercise_boundary_row[
                    "candidate_core_record_count"
                ],

            "candidate_core_token_count":
                failed_exercise_boundary_row[
                    "candidate_core_token_count"
                ],

            "selected_positions_inside_core":
                failed_exercise_boundary_row[
                    "selected_case_positions_inside_candidate_core"
                ],

            "selected_positions_after_boundary":
                failed_exercise_boundary_row[
                    "selected_case_positions_after_candidate_boundary"
                ],
        },
        ensure_ascii=False,
        sort_keys=True,
    )
)


print(
    "\n=== CANDIDATE CORE INTEGRITY FAILURES ==="
)

print(
    "Failures:",
    len(
        candidate_core_integrity_failures
    )
)


if (
    candidate_core_integrity_failures
):

    for failure in (
        candidate_core_integrity_failures
    ):

        print(
            "-",
            failure
        )


print(
    "\n=== CODE CELL 35 GATES ==="
)

for name, passed in (
    cell_35_gates.items()
):

    print(
        name,
        ":",
        "PASS"
        if passed
        else "FAIL"
    )


print(
    "\nCODE CELL 35 GATE:",
    "PASS"
    if cell_35_pass
    else "FAIL"
)


assert cell_35_pass, (
    "Exercise boundary candidate discovery failed. "
    "Do not repair exercise Golden Case identity, evidence scope, "
    "or Target Evidence until explicit and unresolved exercise "
    "boundaries are separately understood."
)
```

<h3>Full Inspection of the Six Unresolved Exercise Groups</h3>

<p>
Cell 35 برای ۱۲ Exercise Group یک Boundary Candidate صریح پیدا کرد،
اما ۶ Group هنوز هیچ Marker صریحی مانند
<code>مطالعه بیشتر</code>
یا
<code>منابع بیشتر</code>
ندارند.

برای این ۶ Group نباید از URL، پایان Table، Token Count یا موقعیت Candidateهای قبلی
به‌عنوان Boundary حدسی استفاده شود.
</p>

<p>
در این مرحله تمام Recordهای این ۶ Group همراه با Structural Class،
Table State، URL-like State، Cueها و Selected-record positions نمایش داده می‌شوند.

<p>
هدف فقط تعیین ورودی لازم برای مرحله‌ی بعد است:
آیا کل Heading Group می‌تواند به‌عنوان یک Exercise Unit معتبر در نظر گرفته شود،
یا داخل Group شواهدی از Supplemental Material وجود دارد که بدون Marker صریح
به Review بیشتری نیاز دارد؟
</p>

<p>
هیچ Boundary نهایی، Case Identity جدید، Evidence Expansion،
Target Repair یا Question Generation در این Cell انجام نمی‌شود.
</p>

```python
# Code Cell 36

# ============================================================
# Full Inspection of Unresolved Exercise Groups
#
# INSPECTION ONLY.
#
# NO fallback boundary.
# NO final boundary freeze.
# NO Golden Case merge/delete.
# NO evidence mutation.
# NO target mutation.
# NO question generation.
# NO expected-answer generation.
# NO artifact writes.
# ============================================================


# ============================================================
# 1) Resolve the six unresolved groups exactly
# ============================================================

unresolved_exercise_group_rows = [
    row
    for row
    in exercise_boundary_candidate_rows
    if (
        row[
            "boundary_status"
        ]
        ==
        "NO_EXPLICIT_BOUNDARY_CANDIDATE"
    )
]


assert (
    len(
        unresolved_exercise_group_rows
    )
    == 6
)


# ============================================================
# 2) Supplemental-looking descriptive signals
#
# These are NOT boundary rules.
#
# Purpose:
# expose possible structural transitions that may require
# review even when no explicit "مطالعه بیشتر / منابع بیشتر"
# marker exists.
# ============================================================

SUPPLEMENTAL_DESCRIPTIVE_CUES = [
    "منبع",
    "منابع",
    "مطالعه",
    "بیشتر بخوان",
    "بیشتر بدان",
    "برای آشنایی",
    "برای اطلاعات",
    "سامانه",
    "سایت",
    "وب سایت",
    "وب‌سایت",
    "اپلیکیشن",
    "برنامه کاربردی",
    "کتاب",
    "مقاله",
    "لینک",
]


def supplemental_descriptive_hits(
    text,
):

    normalized = (
        normalized_boundary_text(
            text
        )
    )

    return [
        cue
        for cue
        in SUPPLEMENTAL_DESCRIPTIVE_CUES
        if (
            normalized_boundary_text(
                cue
            )
            in normalized
        )
    ]


# ============================================================
# 3) Build full inspection rows
# ============================================================

unresolved_full_inspection_rows = []


for unresolved_row in (
    unresolved_exercise_group_rows
):

    group_key = (
        unresolved_row[
            "structure_group_key"
        ]
    )

    group_records = (
        book_heading_groups[
            group_key
        ]
    )


    selected_document_ids = set()

    corresponding_group_discovery = next(
        row
        for row
        in exercise_group_boundary_discovery_rows
        if (
            row[
                "structure_group_key"
            ]
            ==
            group_key
        )
    )


    selected_document_ids.update(
        corresponding_group_discovery[
            "selected_candidate_document_ids"
        ]
    )


    full_sequence = []


    for index, record in enumerate(
        group_records
    ):

        exercise_hits = (
            descriptive_cue_hits(
                record[
                    "text"
                ]
            )
        )

        supplemental_hits = (
            supplemental_descriptive_hits(
                record[
                    "text"
                ]
            )
        )

        url_like = (
            is_url_like_text(
                record[
                    "text"
                ]
            )
        )


        full_sequence.append(
            {
                "index":
                    index,

                "paragraph_number":
                    record[
                        "paragraph_number"
                    ],

                "document_id":
                    record[
                        "document_id"
                    ],

                "record_class":
                    structural_record_class(
                        record
                    ),

                "is_heading":
                    record[
                        "is_heading"
                    ],

                "in_table":
                    record[
                        "in_table"
                    ],

                "token_count":
                    record[
                        "token_count"
                    ],

                "url_like":
                    url_like,

                "exercise_cue_hits":
                    exercise_hits,

                "supplemental_descriptive_hits":
                    supplemental_hits,

                "selected_as_previous_golden_candidate":
                    (
                        record[
                            "document_id"
                        ]
                        in
                        selected_document_ids
                    ),

                "text":
                    record[
                        "text"
                    ],
            }
        )


    # --------------------------------------------------------
    # Structural run summary
    # --------------------------------------------------------

    structural_runs = (
        build_structural_runs(
            group_records
        )
    )


    # --------------------------------------------------------
    # Descriptive counts only
    # --------------------------------------------------------

    supplemental_signal_records = [
        item
        for item
        in full_sequence
        if (
            item[
                "url_like"
            ]

            or

            bool(
                item[
                    "supplemental_descriptive_hits"
                ]
            )
        )
    ]


    exercise_signal_records = [
        item
        for item
        in full_sequence
        if (
            bool(
                item[
                    "exercise_cue_hits"
                ]
            )
        )
    ]


    selected_positions = [
        item[
            "index"
        ]
        for item
        in full_sequence
        if (
            item[
                "selected_as_previous_golden_candidate"
            ]
        )
    ]


    unresolved_full_inspection_rows.append(
        {
            "work_id":
                corresponding_group_discovery[
                    "work_id"
                ],

            "edition_id":
                corresponding_group_discovery[
                    "edition_id"
                ],

            "heading_path":
                corresponding_group_discovery[
                    "heading_path"
                ],

            "structure_group_key":
                group_key,

            "group_record_count":
                len(
                    group_records
                ),

            "group_total_tokens":
                sum(
                    record[
                        "token_count"
                    ]
                    for record
                    in group_records
                ),

            "selected_golden_case_count":
                corresponding_group_discovery[
                    "selected_golden_case_count"
                ],

            "selected_candidate_positions":
                selected_positions,

            "table_record_count":
                sum(
                    1
                    for record
                    in group_records
                    if (
                        record[
                            "in_table"
                        ]
                        is True
                    )
                ),

            "url_like_record_count":
                sum(
                    1
                    for item
                    in full_sequence
                    if (
                        item[
                            "url_like"
                        ]
                    )
                ),

            "exercise_signal_record_count":
                len(
                    exercise_signal_records
                ),

            "supplemental_signal_record_count":
                len(
                    supplemental_signal_records
                ),

            "exercise_signal_records":
                exercise_signal_records,

            "supplemental_signal_records":
                supplemental_signal_records,

            "structural_runs":
                structural_runs,

            "full_sequence":
                full_sequence,

            "boundary_resolution_status":
                "NOT_YET_DECIDED",
        }
    )


# ============================================================
# 4) Descriptive classification of evidence complexity
#
# This is NOT a final boundary decision.
# ============================================================

for row in (
    unresolved_full_inspection_rows
):

    has_url_like = (
        row[
            "url_like_record_count"
        ]
        > 0
    )

    has_supplemental_text_signal = any(
        bool(
            item[
                "supplemental_descriptive_hits"
            ]
        )
        for item
        in row[
            "full_sequence"
        ]
    )

    has_table = (
        row[
            "table_record_count"
        ]
        > 0
    )


    if (
        not has_url_like

        and

        not has_supplemental_text_signal
    ):

        descriptive_complexity = (
            "NO_OBSERVED_SUPPLEMENTAL_SIGNAL"
        )


    elif (
        has_url_like

        or

        has_supplemental_text_signal
    ):

        descriptive_complexity = (
            "POSSIBLE_SUPPLEMENTAL_SIGNAL_PRESENT"
        )


    else:

        descriptive_complexity = (
            "UNCLASSIFIED"
        )


    row[
        "descriptive_complexity"
    ] = (
        descriptive_complexity
    )

    row[
        "has_table"
    ] = (
        has_table
    )


# ============================================================
# 5) Aggregate unresolved-group evidence
# ============================================================

unresolved_complexity_counts = Counter(
    row[
        "descriptive_complexity"
    ]
    for row
    in unresolved_full_inspection_rows
)


unresolved_groups_without_observed_supplemental_signal = [
    row
    for row
    in unresolved_full_inspection_rows
    if (
        row[
            "descriptive_complexity"
        ]
        ==
        "NO_OBSERVED_SUPPLEMENTAL_SIGNAL"
    )
]


unresolved_groups_with_possible_supplemental_signal = [
    row
    for row
    in unresolved_full_inspection_rows
    if (
        row[
            "descriptive_complexity"
        ]
        ==
        "POSSIBLE_SUPPLEMENTAL_SIGNAL_PRESENT"
    )
]


# ============================================================
# 6) Selected-case reuse inside unresolved groups
# ============================================================

unresolved_selected_case_count = sum(
    row[
        "selected_golden_case_count"
    ]
    for row
    in unresolved_full_inspection_rows
)


unresolved_groups_with_multiple_selected_cases = [
    row
    for row
    in unresolved_full_inspection_rows
    if (
        row[
            "selected_golden_case_count"
        ]
        > 1
    )
]


# ============================================================
# 7) Independent source integrity
# ============================================================

unresolved_record_integrity_failures = []


for row in (
    unresolved_full_inspection_rows
):

    for item in (
        row[
            "full_sequence"
        ]
    ):

        document_id = (
            item[
                "document_id"
            ]
        )

        record = (
            corpus_target_index[
                document_id
            ]
        )


        if not (
            record[
                "text"
            ]
            or ""
        ).strip():

            unresolved_record_integrity_failures.append(
                {
                    "document_id":
                        document_id,

                    "reason":
                        "EMPTY_TEXT",
                }
            )


        if not (
            record[
                "locator"
            ]
        ):

            unresolved_record_integrity_failures.append(
                {
                    "document_id":
                        document_id,

                    "reason":
                        "MISSING_LOCATOR",
                }
            )


        if (
            record[
                "lineage"
            ]
            is None
        ):

            unresolved_record_integrity_failures.append(
                {
                    "document_id":
                        document_id,

                    "reason":
                        "MISSING_LINEAGE",
                }
            )


# ============================================================
# 8) Inspection registry
# ============================================================

UNRESOLVED_EXERCISE_GROUP_INSPECTION_V1_DRAFT = {
    "version":
        "unresolved_exercise_group_inspection_v1_draft",

    "status":
        "FULL_SEQUENCE_EXPOSED_NO_BOUNDARY_DECISION",

    "unresolved_group_count":
        len(
            unresolved_full_inspection_rows
        ),

    "selected_record_level_case_count":
        unresolved_selected_case_count,

    "groups_without_observed_supplemental_signal":
        len(
            unresolved_groups_without_observed_supplemental_signal
        ),

    "groups_with_possible_supplemental_signal":
        len(
            unresolved_groups_with_possible_supplemental_signal
        ),

    "fallback_boundary_policy":
        "NONE",

    "full_group_auto_acceptance":
        False,

    "boundary_freeze_status":
        "NOT_STARTED",

    "case_identity_repair_status":
        "NOT_STARTED",

    "evidence_scope_repair_status":
        "NOT_STARTED",

    "target_repair_status":
        "NOT_STARTED",

    "question_generation_status":
        "NOT_STARTED",

    "expected_answer_generation_status":
        "NOT_STARTED",

    "artifact_write_status":
        "NONE",
}


# ============================================================
# 9) Validation gates
# ============================================================

exact_6_unresolved_groups_gate = (
    len(
        unresolved_full_inspection_rows
    )
    == 6
)


all_unresolved_sequences_complete_gate = all(
    len(
        row[
            "full_sequence"
        ]
    )
    ==
    row[
        "group_record_count"
    ]

    for row
    in unresolved_full_inspection_rows
)


all_unresolved_structural_runs_present_gate = all(
    len(
        row[
            "structural_runs"
        ]
    )
    >= 1

    for row
    in unresolved_full_inspection_rows
)


all_unresolved_status_undecided_gate = all(
    row[
        "boundary_resolution_status"
    ]
    ==
    "NOT_YET_DECIDED"

    for row
    in unresolved_full_inspection_rows
)


unresolved_record_integrity_gate = (
    len(
        unresolved_record_integrity_failures
    )
    == 0
)


no_fallback_boundary_gate = (
    UNRESOLVED_EXERCISE_GROUP_INSPECTION_V1_DRAFT[
        "fallback_boundary_policy"
    ]
    ==
    "NONE"
)


no_full_group_auto_acceptance_gate = (
    UNRESOLVED_EXERCISE_GROUP_INSPECTION_V1_DRAFT[
        "full_group_auto_acceptance"
    ]
    is False
)


boundary_freeze_not_started_gate = (
    UNRESOLVED_EXERCISE_GROUP_INSPECTION_V1_DRAFT[
        "boundary_freeze_status"
    ]
    ==
    "NOT_STARTED"
)


case_identity_repair_not_started_gate = (
    UNRESOLVED_EXERCISE_GROUP_INSPECTION_V1_DRAFT[
        "case_identity_repair_status"
    ]
    ==
    "NOT_STARTED"
)


evidence_scope_repair_not_started_gate = (
    UNRESOLVED_EXERCISE_GROUP_INSPECTION_V1_DRAFT[
        "evidence_scope_repair_status"
    ]
    ==
    "NOT_STARTED"
)


target_repair_not_started_gate = (
    UNRESOLVED_EXERCISE_GROUP_INSPECTION_V1_DRAFT[
        "target_repair_status"
    ]
    ==
    "NOT_STARTED"
)


question_generation_not_started_gate = (
    UNRESOLVED_EXERCISE_GROUP_INSPECTION_V1_DRAFT[
        "question_generation_status"
    ]
    ==
    "NOT_STARTED"
)


expected_answer_not_started_gate = (
    UNRESOLVED_EXERCISE_GROUP_INSPECTION_V1_DRAFT[
        "expected_answer_generation_status"
    ]
    ==
    "NOT_STARTED"
)


no_artifact_write_gate = (
    UNRESOLVED_EXERCISE_GROUP_INSPECTION_V1_DRAFT[
        "artifact_write_status"
    ]
    ==
    "NONE"
)


cell_36_gates = {
    "exact_6_unresolved_exercise_groups_fully_inspected":
        exact_6_unresolved_groups_gate,

    "all_unresolved_record_sequences_complete":
        all_unresolved_sequences_complete_gate,

    "all_unresolved_groups_have_structural_runs":
        all_unresolved_structural_runs_present_gate,

    "all_unresolved_boundary_decisions_remain_undecided":
        all_unresolved_status_undecided_gate,

    "all_unresolved_records_preserve_text_locator_lineage":
        unresolved_record_integrity_gate,

    "no_fallback_boundary_invented":
        no_fallback_boundary_gate,

    "no_full_group_auto_acceptance":
        no_full_group_auto_acceptance_gate,

    "boundary_freeze_not_started":
        boundary_freeze_not_started_gate,

    "case_identity_repair_not_started":
        case_identity_repair_not_started_gate,

    "evidence_scope_repair_not_started":
        evidence_scope_repair_not_started_gate,

    "target_repair_not_started":
        target_repair_not_started_gate,

    "question_generation_not_started":
        question_generation_not_started_gate,

    "expected_answer_generation_not_started":
        expected_answer_not_started_gate,

    "no_artifact_write":
        no_artifact_write_gate,
}


cell_36_pass = all(
    cell_36_gates.values()
)


# ============================================================
# 10) Output
# ============================================================

print(
    "=== FULL UNRESOLVED EXERCISE GROUP INSPECTION ==="
)

print(
    "Groups inspected:",
    len(
        unresolved_full_inspection_rows
    )
)

print(
    "Record-level selected cases inside these groups:",
    unresolved_selected_case_count
)


print(
    "\n=== DESCRIPTIVE COMPLEXITY ==="
)

for status, count in sorted(
    unresolved_complexity_counts.items()
):

    print(
        "-",
        status,
        ":",
        count,
    )


print(
    "\nGroups without observed supplemental signal:",
    len(
        unresolved_groups_without_observed_supplemental_signal
    )
)

print(
    "Groups with possible supplemental signal:",
    len(
        unresolved_groups_with_possible_supplemental_signal
    )
)


print(
    "\n=== UNRESOLVED GROUPS WITH MULTIPLE SELECTED CASES ==="
)

for row in (
    unresolved_groups_with_multiple_selected_cases
):

    print(
        "-",
        row[
            "heading_path"
        ],
        "| selected cases:",
        row[
            "selected_golden_case_count"
        ],
        "| positions:",
        row[
            "selected_candidate_positions"
        ],
    )


print(
    "\n=== COMPLETE UNRESOLVED GROUP RECORDS ==="
)


for group_index, row in enumerate(
    unresolved_full_inspection_rows,
    start=1,
):

    print(
        "\n"
        + "=" * 100
    )

    print(
        "UNRESOLVED GROUP",
        group_index,
        "/",
        len(
            unresolved_full_inspection_rows
        )
    )

    print(
        "Heading path:",
        row[
            "heading_path"
        ]
    )

    print(
        "Records:",
        row[
            "group_record_count"
        ],
        "| tokens:",
        row[
            "group_total_tokens"
        ],
        "| selected cases:",
        row[
            "selected_golden_case_count"
        ],
    )

    print(
        "Selected positions:",
        row[
            "selected_candidate_positions"
        ]
    )

    print(
        "Table records:",
        row[
            "table_record_count"
        ]
    )

    print(
        "URL-like records:",
        row[
            "url_like_record_count"
        ]
    )

    print(
        "Exercise-signal records:",
        row[
            "exercise_signal_record_count"
        ]
    )

    print(
        "Supplemental descriptive-signal records:",
        row[
            "supplemental_signal_record_count"
        ]
    )

    print(
        "Descriptive complexity:",
        row[
            "descriptive_complexity"
        ]
    )


    print(
        "\nStructural runs:"
    )

    for run in (
        row[
            "structural_runs"
        ]
    ):

        print(
            json.dumps(
                {
                    "record_class":
                        run[
                            "record_class"
                        ],

                    "start_index":
                        run[
                            "start_index"
                        ],

                    "end_index":
                        run[
                            "end_index"
                        ],

                    "record_count":
                        run[
                            "record_count"
                        ],

                    "token_count":
                        run[
                            "token_count"
                        ],
                },
                ensure_ascii=False,
                sort_keys=True,
            )
        )


    print(
        "\nFull sequence:"
    )


    for item in (
        row[
            "full_sequence"
        ]
    ):

        print(
            "- index:",
            item[
                "index"
            ],
            "| paragraph:",
            item[
                "paragraph_number"
            ],
            "| class:",
            item[
                "record_class"
            ],
            "| selected:",
            item[
                "selected_as_previous_golden_candidate"
            ],
            "| exercise cues:",
            item[
                "exercise_cue_hits"
            ],
            "| supplemental cues:",
            item[
                "supplemental_descriptive_hits"
            ],
            "| URL:",
            item[
                "url_like"
            ],
            "| tokens:",
            item[
                "token_count"
            ],
        )

        print(
            "  TEXT:",
            item[
                "text"
            ]
        )


    print(
        "\nBOUNDARY RESOLUTION:",
        row[
            "boundary_resolution_status"
        ]
    )


print(
    "\n=== RECORD INTEGRITY FAILURES ==="
)

print(
    "Failures:",
    len(
        unresolved_record_integrity_failures
    )
)


if (
    unresolved_record_integrity_failures
):

    for failure in (
        unresolved_record_integrity_failures[
            :20
        ]
    ):

        print(
            "-",
            failure
        )


print(
    "\n=== CODE CELL 36 GATES ==="
)

for name, passed in (
    cell_36_gates.items()
):

    print(
        name,
        ":",
        "PASS"
        if passed
        else "FAIL"
    )


print(
    "\nCODE CELL 36 GATE:",
    "PASS"
    if cell_36_pass
    else "FAIL"
)


assert cell_36_pass, (
    "Unresolved exercise-group inspection failed. "
    "Do not freeze exercise boundaries or repair Golden Case "
    "identity until all six unresolved groups are fully exposed."
)
```

```python
# Code Cell 37

# ============================================================
# Adjudication of the 6 Unresolved Exercise Boundaries
#
# Evidence-based authoring decisions using the complete
# sequences exposed in Cell 36.
#
# NO Golden Case merge/delete.
# NO Case ID mutation.
# NO evidence-scope mutation.
# NO target mutation.
# NO question generation.
# NO expected-answer generation.
# NO artifact writes.
# ============================================================


# ============================================================
# 1) Explicit adjudication contract
#
# Boundary indices below refer to the exact full_sequence
# indices exposed in Cell 36.
# ============================================================

UNRESOLVED_EXERCISE_BOUNDARY_ADJUDICATION = {

    (
        "book:adult-financial-literacy-course",
        "book:adult-financial-literacy-course:fa-source",
        (
            "درس ۲: پلکان توانمندی مالی",
            "خودارزیابی 1: موقعیت من در پلکان توانمندی مالی",
        ),
    ): {
        "decision":
            "FULL_GROUP_IS_EXERCISE_UNIT",

        "exercise_core_start_index":
            0,

        "exercise_core_end_index":
            2,

        "first_nonexercise_index":
            None,

        "basis":
            (
                "Heading plus both following narrative records form one "
                "continuous self-assessment instruction; no transition "
                "to supplemental or lesson content is present."
            ),
    },


    (
        "book:adult-financial-literacy-course",
        "book:adult-financial-literacy-course:fa-source",
        (
            "درس ۲۲: صندوق‌های سرمایه‌گذاری",
            "خودارزیابی ۱۵: کدام صندوق برای من",
        ),
    ): {
        "decision":
            "FULL_GROUP_IS_EXERCISE_UNIT",

        "exercise_core_start_index":
            0,

        "exercise_core_end_index":
            152,

        "first_nonexercise_index":
            None,

        "basis":
            (
                "The group is one questionnaire: introduction, question/"
                "option/score table records, and final scoring analysis. "
                "No supplemental or unrelated lesson transition appears."
            ),
    },


    (
        "book:adult-financial-literacy-course",
        "book:adult-financial-literacy-course:fa-source",
        (
            "درس ۲۵: مؤلفه‌های مالی در تصمیم‌گیری مالی",
            "خودارزیابی ۱۹: ذهن ثروتمند یا فقیر من",
        ),
    ): {
        "decision":
            "INTERNAL_CONTENT_TRANSITION_BOUNDARY",

        "exercise_core_start_index":
            0,

        "exercise_core_end_index":
            1,

        "first_nonexercise_index":
            2,

        "basis":
            (
                "Index 1 is the self-assessment step. Index 2 explicitly "
                "starts a new conceptual section: "
                "'۲. شاخص دوم تصمیم‌گیری مالی:انواع تمایلات'."
            ),
    },


    (
        "book:adult-financial-literacy-course",
        "book:adult-financial-literacy-course:fa-source",
        (
            "درس ۲۵: مؤلفه‌های مالی در تصمیم‌گیری مالی",
            "کاربرگ 9: فهرست دارایی‌ها و صد میلیون هدیه",
        ),
    ): {
        "decision":
            "INTERNAL_CONTENT_TRANSITION_BOUNDARY",

        "exercise_core_start_index":
            0,

        "exercise_core_end_index":
            6,

        "first_nonexercise_index":
            7,

        "basis":
            (
                "Indices 0-6 contain the worksheet heading, two explicit "
                "steps and worksheet table structure. Index 7 begins "
                "new lesson exposition: "
                "'سه مؤلفه مهم در تصمیم‌گیری وجود دارد:'."
            ),
    },


    (
        "book:trainer-training-financial-literacy",
        "book:trainer-training-financial-literacy:fa-source",
        (
            "مبادلۀ کالابه‌کالا",
            "خودارزیابی 1",
        ),
    ): {
        "decision":
            "FULL_GROUP_IS_EXERCISE_UNIT",

        "exercise_core_start_index":
            0,

        "exercise_core_end_index":
            3,

        "first_nonexercise_index":
            None,

        "basis":
            (
                "All three narrative records after the heading are "
                "self-assessment questions about prior and revised "
                "financial decisions; no content transition is present."
            ),
    },


    (
        "book:trainer-training-financial-literacy",
        "book:trainer-training-financial-literacy:fa-source",
        (
            "پس‌انداز",
            "خودارزیابی 2",
        ),
    ): {
        "decision":
            "FULL_GROUP_IS_EXERCISE_UNIT",

        "exercise_core_start_index":
            0,

        "exercise_core_end_index":
            8,

        "first_nonexercise_index":
            None,

        "basis":
            (
                "All eight narrative records after the heading are a "
                "continuous sequence of self-assessment questions about "
                "financial decisions, delayed gratification and saving."
            ),
    },
}


# ============================================================
# 2) Resolve adjudication against actual Cell 36 groups
# ============================================================

unresolved_boundary_adjudication_rows = []


for row in (
    unresolved_full_inspection_rows
):

    key = (
        row[
            "work_id"
        ],
        row[
            "edition_id"
        ],
        tuple(
            row[
                "heading_path"
            ]
        ),
    )


    decision = (
        UNRESOLVED_EXERCISE_BOUNDARY_ADJUDICATION.get(
            key
        )
    )


    if (
        decision
        is None
    ):

        raise AssertionError(
            "Missing unresolved boundary adjudication for: "
            f"{key}"
        )


    start_index = (
        decision[
            "exercise_core_start_index"
        ]
    )

    end_index = (
        decision[
            "exercise_core_end_index"
        ]
    )

    first_nonexercise_index = (
        decision[
            "first_nonexercise_index"
        ]
    )


    full_sequence = (
        row[
            "full_sequence"
        ]
    )


    assert (
        0
        <= start_index
        <= end_index
        < len(
            full_sequence
        )
    )


    core_items = (
        full_sequence[
            start_index:
            end_index + 1
        ]
    )


    core_document_ids = [
        item[
            "document_id"
        ]
        for item
        in core_items
    ]


    core_token_count = sum(
        item[
            "token_count"
        ]
        for item
        in core_items
    )


    if (
        first_nonexercise_index
        is None
    ):

        nonexercise_items = []

    else:

        nonexercise_items = (
            full_sequence[
                first_nonexercise_index:
            ]
        )


    selected_positions_inside_core = [
        position
        for position
        in row[
            "selected_candidate_positions"
        ]
        if (
            start_index
            <= position
            <= end_index
        )
    ]


    selected_positions_outside_core = [
        position
        for position
        in row[
            "selected_candidate_positions"
        ]
        if not (
            start_index
            <= position
            <= end_index
        )
    ]


    unresolved_boundary_adjudication_rows.append(
        {
            "work_id":
                row[
                    "work_id"
                ],

            "edition_id":
                row[
                    "edition_id"
                ],

            "heading_path":
                row[
                    "heading_path"
                ],

            "structure_group_key":
                row[
                    "structure_group_key"
                ],

            "decision":
                decision[
                    "decision"
                ],

            "basis":
                decision[
                    "basis"
                ],

            "exercise_core_start_index":
                start_index,

            "exercise_core_end_index":
                end_index,

            "first_nonexercise_index":
                first_nonexercise_index,

            "exercise_core_record_count":
                len(
                    core_items
                ),

            "exercise_core_token_count":
                core_token_count,

            "exercise_core_document_ids":
                core_document_ids,

            "nonexercise_record_count":
                len(
                    nonexercise_items
                ),

            "selected_golden_case_count":
                row[
                    "selected_golden_case_count"
                ],

            "selected_candidate_positions":
                row[
                    "selected_candidate_positions"
                ],

            "selected_positions_inside_core":
                selected_positions_inside_core,

            "selected_positions_outside_core":
                selected_positions_outside_core,

            "boundary_adjudication_status":
                "ADJUDICATED_NOT_YET_APPLIED",
        }
    )


# ============================================================
# 3) Independent structural validation
# ============================================================

unresolved_boundary_integrity_failures = []


for row in (
    unresolved_boundary_adjudication_rows
):

    source_group = next(
        source_row
        for source_row
        in unresolved_full_inspection_rows
        if (
            source_row[
                "structure_group_key"
            ]
            ==
            row[
                "structure_group_key"
            ]
        )
    )


    full_sequence = (
        source_group[
            "full_sequence"
        ]
    )


    expected_core_ids = [
        item[
            "document_id"
        ]
        for item
        in full_sequence[
            row[
                "exercise_core_start_index"
            ]:
            row[
                "exercise_core_end_index"
            ] + 1
        ]
    ]


    if (
        expected_core_ids
        !=
        row[
            "exercise_core_document_ids"
        ]
    ):

        unresolved_boundary_integrity_failures.append(
            {
                "heading_path":
                    row[
                        "heading_path"
                    ],

                "reason":
                    "CORE_DOCUMENT_SEQUENCE_MISMATCH",
            }
        )


    if (
        row[
            "decision"
        ]
        ==
        "FULL_GROUP_IS_EXERCISE_UNIT"
    ):

        if not (
            row[
                "exercise_core_start_index"
            ]
            == 0

            and

            row[
                "exercise_core_end_index"
            ]
            ==
            len(
                full_sequence
            )
            - 1

            and

            row[
                "first_nonexercise_index"
            ]
            is None
        ):

            unresolved_boundary_integrity_failures.append(
                {
                    "heading_path":
                        row[
                            "heading_path"
                        ],

                    "reason":
                        "FULL_GROUP_DECISION_NOT_FULL_GROUP",
                }
            )


    if (
        row[
            "decision"
        ]
        ==
        "INTERNAL_CONTENT_TRANSITION_BOUNDARY"
    ):

        if not (
            row[
                "first_nonexercise_index"
            ]
            ==
            row[
                "exercise_core_end_index"
            ]
            + 1
        ):

            unresolved_boundary_integrity_failures.append(
                {
                    "heading_path":
                        row[
                            "heading_path"
                        ],

                    "reason":
                        "INTERNAL_BOUNDARY_NOT_CONTIGUOUS",
                }
            )


# ============================================================
# 4) Combine with 12 explicit-boundary groups
#
# Still a draft boundary contract.
# No Golden Case mutation yet.
# ============================================================

exercise_boundary_contract_rows = []


# ------------------------------------------------------------
# 12 explicit marker-derived groups
# ------------------------------------------------------------

for row in (
    explicit_boundary_rows
):

    exercise_boundary_contract_rows.append(
        {
            "work_id":
                row[
                    "work_id"
                ],

            "edition_id":
                row[
                    "edition_id"
                ],

            "heading_path":
                row[
                    "heading_path"
                ],

            "structure_group_key":
                row[
                    "structure_group_key"
                ],

            "boundary_source":
                "EXPLICIT_SUPPLEMENTAL_MARKER",

            "exercise_core_start_index":
                row[
                    "exercise_core_start_index"
                ],

            "exercise_core_end_index":
                row[
                    "exercise_core_end_index"
                ],

            "first_nonexercise_index":
                row[
                    "supplemental_start_index"
                ],

            "exercise_core_document_ids":
                list(
                    row[
                        "candidate_core_document_ids"
                    ]
                ),

            "exercise_core_record_count":
                row[
                    "candidate_core_record_count"
                ],

            "exercise_core_token_count":
                row[
                    "candidate_core_token_count"
                ],

            "selected_golden_case_count":
                row[
                    "selected_golden_case_count"
                ],

            "selected_positions_inside_core":
                list(
                    row[
                        "selected_case_positions_inside_candidate_core"
                    ]
                ),

            "selected_positions_outside_core":
                list(
                    row[
                        "selected_case_positions_after_candidate_boundary"
                    ]
                ),

            "boundary_status":
                "BOUNDARY_RESOLVED_NOT_APPLIED",
        }
    )


# ------------------------------------------------------------
# 6 adjudicated groups
# ------------------------------------------------------------

for row in (
    unresolved_boundary_adjudication_rows
):

    exercise_boundary_contract_rows.append(
        {
            "work_id":
                row[
                    "work_id"
                ],

            "edition_id":
                row[
                    "edition_id"
                ],

            "heading_path":
                row[
                    "heading_path"
                ],

            "structure_group_key":
                row[
                    "structure_group_key"
                ],

            "boundary_source":
                row[
                    "decision"
                ],

            "exercise_core_start_index":
                row[
                    "exercise_core_start_index"
                ],

            "exercise_core_end_index":
                row[
                    "exercise_core_end_index"
                ],

            "first_nonexercise_index":
                row[
                    "first_nonexercise_index"
                ],

            "exercise_core_document_ids":
                list(
                    row[
                        "exercise_core_document_ids"
                    ]
                ),

            "exercise_core_record_count":
                row[
                    "exercise_core_record_count"
                ],

            "exercise_core_token_count":
                row[
                    "exercise_core_token_count"
                ],

            "selected_golden_case_count":
                row[
                    "selected_golden_case_count"
                ],

            "selected_positions_inside_core":
                list(
                    row[
                        "selected_positions_inside_core"
                    ]
                ),

            "selected_positions_outside_core":
                list(
                    row[
                        "selected_positions_outside_core"
                    ]
                ),

            "boundary_status":
                "BOUNDARY_RESOLVED_NOT_APPLIED",
        }
    )


# ============================================================
# 5) Aggregate final 18-group boundary evidence
# ============================================================

boundary_source_counts = Counter(
    row[
        "boundary_source"
    ]
    for row
    in exercise_boundary_contract_rows
)


selected_record_cases_inside_resolved_cores = sum(
    len(
        row[
            "selected_positions_inside_core"
        ]
    )
    for row
    in exercise_boundary_contract_rows
)


selected_record_cases_outside_resolved_cores = sum(
    len(
        row[
            "selected_positions_outside_core"
        ]
    )
    for row
    in exercise_boundary_contract_rows
)


groups_with_record_candidates_outside_core = [
    row
    for row
    in exercise_boundary_contract_rows
    if (
        len(
            row[
                "selected_positions_outside_core"
            ]
        )
        > 0
    )
]


exercise_core_record_count_values = [
    row[
        "exercise_core_record_count"
    ]
    for row
    in exercise_boundary_contract_rows
]


exercise_core_token_count_values = [
    row[
        "exercise_core_token_count"
    ]
    for row
    in exercise_boundary_contract_rows
]


# ============================================================
# 6) Boundary Contract Registry
# ============================================================

EXERCISE_BOUNDARY_CONTRACT_V1_DRAFT = {
    "version":
        "exercise_boundary_contract_v1_draft",

    "status":
        "ALL_18_BOUNDARIES_RESOLVED_NOT_APPLIED",

    "exercise_group_count":
        len(
            exercise_boundary_contract_rows
        ),

    "explicit_marker_boundary_count":
        boundary_source_counts[
            "EXPLICIT_SUPPLEMENTAL_MARKER"
        ],

    "full_group_exercise_count":
        boundary_source_counts[
            "FULL_GROUP_IS_EXERCISE_UNIT"
        ],

    "internal_content_transition_boundary_count":
        boundary_source_counts[
            "INTERNAL_CONTENT_TRANSITION_BOUNDARY"
        ],

    "record_level_selected_case_count":
        38,

    "selected_record_cases_inside_resolved_exercise_core":
        selected_record_cases_inside_resolved_cores,

    "selected_record_cases_outside_resolved_exercise_core":
        selected_record_cases_outside_resolved_cores,

    "case_identity_repair_status":
        "NOT_STARTED",

    "evidence_scope_repair_status":
        "NOT_STARTED",

    "target_repair_status":
        "NOT_STARTED",

    "question_generation_status":
        "NOT_STARTED",

    "expected_answer_generation_status":
        "NOT_STARTED",

    "artifact_write_status":
        "NONE",
}


# ============================================================
# 7) Validation gates
# ============================================================

exact_6_unresolved_decisions_gate = (
    len(
        unresolved_boundary_adjudication_rows
    )
    == 6
)


four_full_group_decisions_gate = (
    sum(
        1
        for row
        in unresolved_boundary_adjudication_rows
        if (
            row[
                "decision"
            ]
            ==
            "FULL_GROUP_IS_EXERCISE_UNIT"
        )
    )
    == 4
)


two_internal_transition_decisions_gate = (
    sum(
        1
        for row
        in unresolved_boundary_adjudication_rows
        if (
            row[
                "decision"
            ]
            ==
            "INTERNAL_CONTENT_TRANSITION_BOUNDARY"
        )
    )
    == 2
)


unresolved_boundary_integrity_gate = (
    len(
        unresolved_boundary_integrity_failures
    )
    == 0
)


exact_18_resolved_boundary_contract_gate = (
    len(
        exercise_boundary_contract_rows
    )
    == 18
)


all_18_structure_groups_unique_gate = (
    len(
        {
            row[
                "structure_group_key"
            ]
            for row
            in exercise_boundary_contract_rows
        }
    )
    == 18
)


all_boundary_cores_nonempty_gate = all(
    row[
        "exercise_core_record_count"
    ]
    >= 1

    and

    row[
        "exercise_core_token_count"
    ]
    > 0

    and

    len(
        row[
            "exercise_core_document_ids"
        ]
    )
    ==
    row[
        "exercise_core_record_count"
    ]

    for row
    in exercise_boundary_contract_rows
)


all_38_previous_selected_positions_accounted_gate = (
    selected_record_cases_inside_resolved_cores
    +
    selected_record_cases_outside_resolved_cores
    ==
    38
)


no_case_identity_repair_gate = (
    EXERCISE_BOUNDARY_CONTRACT_V1_DRAFT[
        "case_identity_repair_status"
    ]
    ==
    "NOT_STARTED"
)


no_evidence_scope_repair_gate = (
    EXERCISE_BOUNDARY_CONTRACT_V1_DRAFT[
        "evidence_scope_repair_status"
    ]
    ==
    "NOT_STARTED"
)


no_target_repair_gate = (
    EXERCISE_BOUNDARY_CONTRACT_V1_DRAFT[
        "target_repair_status"
    ]
    ==
    "NOT_STARTED"
)


question_generation_not_started_gate = (
    EXERCISE_BOUNDARY_CONTRACT_V1_DRAFT[
        "question_generation_status"
    ]
    ==
    "NOT_STARTED"
)


expected_answer_not_started_gate = (
    EXERCISE_BOUNDARY_CONTRACT_V1_DRAFT[
        "expected_answer_generation_status"
    ]
    ==
    "NOT_STARTED"
)


no_artifact_write_gate = (
    EXERCISE_BOUNDARY_CONTRACT_V1_DRAFT[
        "artifact_write_status"
    ]
    ==
    "NONE"
)


cell_37_gates = {
    "exact_6_unresolved_groups_adjudicated":
        exact_6_unresolved_decisions_gate,

    "exact_4_full_group_exercise_decisions":
        four_full_group_decisions_gate,

    "exact_2_internal_content_transition_boundaries":
        two_internal_transition_decisions_gate,

    "all_unresolved_boundary_decisions_structurally_valid":
        unresolved_boundary_integrity_gate,

    "exact_18_exercise_boundaries_resolved":
        exact_18_resolved_boundary_contract_gate,

    "all_18_structural_groups_unique":
        all_18_structure_groups_unique_gate,

    "all_18_exercise_cores_nonempty":
        all_boundary_cores_nonempty_gate,

    "all_38_previous_record_candidates_accounted_for":
        all_38_previous_selected_positions_accounted_gate,

    "case_identity_repair_not_started":
        no_case_identity_repair_gate,

    "evidence_scope_repair_not_started":
        no_evidence_scope_repair_gate,

    "target_repair_not_started":
        no_target_repair_gate,

    "question_generation_not_started":
        question_generation_not_started_gate,

    "expected_answer_generation_not_started":
        expected_answer_not_started_gate,

    "no_artifact_write":
        no_artifact_write_gate,
}


cell_37_pass = all(
    cell_37_gates.values()
)


# ============================================================
# 8) Output
# ============================================================

print(
    "=== UNRESOLVED EXERCISE BOUNDARY ADJUDICATION ==="
)

print(
    "Previously unresolved groups:",
    len(
        unresolved_boundary_adjudication_rows
    )
)

print(
    "Full-group exercise units:",
    sum(
        1
        for row
        in unresolved_boundary_adjudication_rows
        if (
            row[
                "decision"
            ]
            ==
            "FULL_GROUP_IS_EXERCISE_UNIT"
        )
    )
)

print(
    "Internal content-transition boundaries:",
    sum(
        1
        for row
        in unresolved_boundary_adjudication_rows
        if (
            row[
                "decision"
            ]
            ==
            "INTERNAL_CONTENT_TRANSITION_BOUNDARY"
        )
    )
)


print(
    "\n=== SIX ADJUDICATED GROUPS ==="
)

for row in (
    unresolved_boundary_adjudication_rows
):

    print(
        "\nHeading:",
        row[
            "heading_path"
        ]
    )

    print(
        "Decision:",
        row[
            "decision"
        ]
    )

    print(
        "Core indices:",
        (
            row[
                "exercise_core_start_index"
            ],
            row[
                "exercise_core_end_index"
            ],
        )
    )

    print(
        "First non-exercise index:",
        row[
            "first_nonexercise_index"
        ]
    )

    print(
        "Core records:",
        row[
            "exercise_core_record_count"
        ]
    )

    print(
        "Core tokens:",
        row[
            "exercise_core_token_count"
        ]
    )

    print(
        "Selected positions inside core:",
        row[
            "selected_positions_inside_core"
        ]
    )

    print(
        "Selected positions outside core:",
        row[
            "selected_positions_outside_core"
        ]
    )

    print(
        "Basis:",
        row[
            "basis"
        ]
    )


print(
    "\n=== COMPLETE 18-GROUP EXERCISE BOUNDARY CONTRACT ==="
)

print(
    "Resolved exercise groups:",
    len(
        exercise_boundary_contract_rows
    )
)

print(
    "Boundary-source counts:",
    dict(
        boundary_source_counts
    )
)

print(
    "Previous selected record-level cases inside resolved cores:",
    selected_record_cases_inside_resolved_cores
)

print(
    "Previous selected record-level cases outside resolved cores:",
    selected_record_cases_outside_resolved_cores
)

print(
    "Groups containing previous selected records outside core:",
    len(
        groups_with_record_candidates_outside_core
    )
)


print(
    "\nExercise-core record-count distribution:",
    {
        "min":
            min(
                exercise_core_record_count_values
            ),

        "median":
            empirical_nearest_rank(
                exercise_core_record_count_values,
                0.50,
            ),

        "max":
            max(
                exercise_core_record_count_values
            ),
    }
)


print(
    "Exercise-core token-count distribution:",
    {
        "min":
            min(
                exercise_core_token_count_values
            ),

        "median":
            empirical_nearest_rank(
                exercise_core_token_count_values,
                0.50,
            ),

        "max":
            max(
                exercise_core_token_count_values
            ),
    }
)


print(
    "\n=== GROUPS WITH PREVIOUS RECORD-LEVEL CANDIDATES OUTSIDE EXERCISE CORE ==="
)

for row in (
    groups_with_record_candidates_outside_core
):

    print(
        "-",
        row[
            "heading_path"
        ],
        "| inside:",
        row[
            "selected_positions_inside_core"
        ],
        "| outside:",
        row[
            "selected_positions_outside_core"
        ],
    )


print(
    "\n=== BOUNDARY INTEGRITY FAILURES ==="
)

print(
    "Failures:",
    len(
        unresolved_boundary_integrity_failures
    )
)


if (
    unresolved_boundary_integrity_failures
):

    for failure in (
        unresolved_boundary_integrity_failures
    ):

        print(
            "-",
            failure
        )


print(
    "\n=== CODE CELL 37 GATES ==="
)

for name, passed in (
    cell_37_gates.items()
):

    print(
        name,
        ":",
        "PASS"
        if passed
        else "FAIL"
    )


print(
    "\nCODE CELL 37 GATE:",
    "PASS"
    if cell_37_pass
    else "FAIL"
)


assert cell_37_pass, (
    "Exercise boundary adjudication failed. "
    "Do not repair record-level Golden Case identity or evidence "
    "scope until all 18 exercise-unit boundaries are resolved "
    "and all prior selected records are accounted for."
)
```

```python
# Code Cell 38

# ============================================================
# Exercise-Level Case Identity Reconstruction Audit
#
# DESIGN / RECONSTRUCTION IN MEMORY ONLY.
#
# Builds:
#   18 proposed exercise-level case identities
#   mapping from 38 existing record-level Golden Cases
#
# NO existing Golden Case mutation.
# NO case deletion/supersession applied.
# NO global Golden Dataset count change.
# NO Target Evidence mutation.
# NO question generation.
# NO expected-answer generation.
# NO artifact writes.
# ============================================================


# ============================================================
# 1) Proposed exercise-unit case identity contract
# ============================================================

PROPOSED_EXERCISE_UNIT_CASE_SCHEMA_VERSION = (
    "exercise_unit_golden_case_v1_draft"
)


EXERCISE_CASE_IDENTITY_RECONSTRUCTION_CONTRACT_V1_DRAFT = {
    "version":
        "exercise_case_identity_reconstruction_contract_v1_draft",

    "source_type":
        "book",

    "stratum":
        "exercise_or_self_assessment",

    "legacy_identity_level":
        "RECORD_LEVEL",

    "proposed_identity_level":
        "EXERCISE_UNIT_LEVEL",

    "identity_fields": [
        "work_id",
        "edition_id",
        "heading_path",
        "boundary_contract_version",
        "exercise_core_document_ids",
    ],

    "legacy_case_mutation":
        False,

    "global_golden_dataset_mutation":
        False,

    "target_repair_status":
        "NOT_STARTED",

    "question_generation_status":
        "NOT_STARTED",

    "expected_answer_generation_status":
        "NOT_STARTED",

    "artifact_write_status":
        "NONE",
}


# ============================================================
# 2) Stable proposed exercise-unit ID
# ============================================================

def build_proposed_exercise_unit_case_id(
    work_id,
    edition_id,
    heading_path,
    exercise_core_document_ids,
):

    identity_payload = {
        "schema_version":
            PROPOSED_EXERCISE_UNIT_CASE_SCHEMA_VERSION,

        "source_type":
            "book",

        "stratum":
            "exercise_or_self_assessment",

        "work_id":
            work_id,

        "edition_id":
            edition_id,

        "heading_path":
            list(
                heading_path
            ),

        "boundary_contract_version":
            EXERCISE_BOUNDARY_CONTRACT_V1_DRAFT[
                "version"
            ],

        "exercise_core_document_ids":
            list(
                exercise_core_document_ids
            ),
    }


    identity_sha256 = hashlib.sha256(
        stable_json_key(
            identity_payload
        ).encode(
            "utf-8"
        )
    ).hexdigest()


    return (
        "exercise-unit-case:"
        +
        identity_sha256[
            :32
        ]
    )


# ============================================================
# 3) Index the 38 legacy record-level selected cases by
# structural group
# ============================================================

legacy_exercise_cases_by_group = defaultdict(
    list
)


for row in (
    exercise_selected_structure_rows
):

    legacy_exercise_cases_by_group[
        row[
            "structure_group_key"
        ]
    ].append(
        row
    )


for group_key in list(
    legacy_exercise_cases_by_group
):

    legacy_exercise_cases_by_group[
        group_key
    ] = sorted(
        legacy_exercise_cases_by_group[
            group_key
        ],
        key=lambda row: (
            row[
                "candidate_position"
            ],
            row[
                "golden_case_id"
            ],
        ),
    )


# ============================================================
# 4) Build 18 proposed exercise-unit cases
# ============================================================

proposed_exercise_unit_cases = []

proposed_exercise_unit_cases_by_id = {}

proposed_exercise_unit_cases_by_group = {}


for boundary_row in sorted(
    exercise_boundary_contract_rows,
    key=lambda row: (
        str(
            row[
                "work_id"
            ]
        ),
        str(
            row[
                "edition_id"
            ]
        ),
        tuple(
            row[
                "heading_path"
            ]
        ),
    ),
):

    group_key = (
        boundary_row[
            "structure_group_key"
        ]
    )


    core_document_ids = list(
        boundary_row[
            "exercise_core_document_ids"
        ]
    )


    legacy_rows = (
        legacy_exercise_cases_by_group[
            group_key
        ]
    )


    legacy_case_ids = [
        row[
            "golden_case_id"
        ]
        for row
        in legacy_rows
    ]


    legacy_candidate_document_ids = [
        row[
            "candidate_document_id"
        ]
        for row
        in legacy_rows
    ]


    core_document_id_set = set(
        core_document_ids
    )


    legacy_cases_inside_core = [
        row[
            "golden_case_id"
        ]
        for row
        in legacy_rows
        if (
            row[
                "candidate_document_id"
            ]
            in
            core_document_id_set
        )
    ]


    legacy_cases_outside_core = [
        row[
            "golden_case_id"
        ]
        for row
        in legacy_rows
        if (
            row[
                "candidate_document_id"
            ]
            not in
            core_document_id_set
        )
    ]


    # --------------------------------------------------------
    # Evidence content fingerprint
    #
    # Uses exact ordered core records.
    # --------------------------------------------------------

    evidence_fingerprint_payload = []


    for document_id in (
        core_document_ids
    ):

        record = (
            corpus_target_index[
                document_id
            ]
        )


        text_sha256 = hashlib.sha256(
            record[
                "text"
            ].encode(
                "utf-8"
            )
        ).hexdigest()


        evidence_fingerprint_payload.append(
            {
                "document_id":
                    document_id,

                "text_sha256":
                    text_sha256,
            }
        )


    exercise_core_evidence_sha256 = hashlib.sha256(
        stable_json_key(
            evidence_fingerprint_payload
        ).encode(
            "utf-8"
        )
    ).hexdigest()


    proposed_case_id = (
        build_proposed_exercise_unit_case_id(
            work_id=(
                boundary_row[
                    "work_id"
                ]
            ),
            edition_id=(
                boundary_row[
                    "edition_id"
                ]
            ),
            heading_path=(
                boundary_row[
                    "heading_path"
                ]
            ),
            exercise_core_document_ids=(
                core_document_ids
            ),
        )
    )


    proposed_record = {
        "proposed_exercise_unit_case_id":
            proposed_case_id,

        "case_schema_version":
            PROPOSED_EXERCISE_UNIT_CASE_SCHEMA_VERSION,

        "source_type":
            "book",

        "stratum":
            "exercise_or_self_assessment",

        "identity_level":
            "EXERCISE_UNIT",

        "work_id":
            boundary_row[
                "work_id"
            ],

        "edition_id":
            boundary_row[
                "edition_id"
            ],

        "heading_path":
            list(
                boundary_row[
                    "heading_path"
                ]
            ),

        "boundary_source":
            boundary_row[
                "boundary_source"
            ],

        "exercise_core_start_index":
            boundary_row[
                "exercise_core_start_index"
            ],

        "exercise_core_end_index":
            boundary_row[
                "exercise_core_end_index"
            ],

        "first_nonexercise_index":
            boundary_row[
                "first_nonexercise_index"
            ],

        "evidence_document_ids":
            core_document_ids,

        "evidence_document_count":
            len(
                core_document_ids
            ),

        "exercise_core_token_count":
            boundary_row[
                "exercise_core_token_count"
            ],

        "exercise_core_evidence_sha256":
            exercise_core_evidence_sha256,

        "legacy_record_level_case_ids":
            legacy_case_ids,

        "legacy_record_level_case_count":
            len(
                legacy_case_ids
            ),

        "legacy_candidate_document_ids":
            legacy_candidate_document_ids,

        "legacy_cases_inside_exercise_core":
            legacy_cases_inside_core,

        "legacy_cases_outside_exercise_core":
            legacy_cases_outside_core,

        "legacy_inside_core_count":
            len(
                legacy_cases_inside_core
            ),

        "legacy_outside_core_count":
            len(
                legacy_cases_outside_core
            ),

        "proposal_status":
            "PROPOSED_NOT_APPLIED",

        "final_golden_membership":
            False,

        "supersession_status":
            "NOT_APPLIED",

        "target_evidence_status":
            "NOT_STARTED",

        "question_generation_status":
            "NOT_STARTED",

        "expected_answer_generation_status":
            "NOT_STARTED",
    }


    proposed_exercise_unit_cases.append(
        proposed_record
    )


    if (
        proposed_case_id
        in
        proposed_exercise_unit_cases_by_id
    ):

        raise AssertionError(
            "Duplicate proposed exercise-unit case ID: "
            f"{proposed_case_id}"
        )


    proposed_exercise_unit_cases_by_id[
        proposed_case_id
    ] = (
        proposed_record
    )


    if (
        group_key
        in
        proposed_exercise_unit_cases_by_group
    ):

        raise AssertionError(
            "Duplicate structural group in proposed cases: "
            f"{group_key}"
        )


    proposed_exercise_unit_cases_by_group[
        group_key
    ] = (
        proposed_record
    )


# ============================================================
# 5) Build exact mapping:
#
# 38 legacy record-level case IDs
#      ->
# 18 proposed exercise-unit case IDs
# ============================================================

legacy_to_proposed_exercise_case_mapping = []


for legacy_row in (
    exercise_selected_structure_rows
):

    group_key = (
        legacy_row[
            "structure_group_key"
        ]
    )


    proposed_case = (
        proposed_exercise_unit_cases_by_group[
            group_key
        ]
    )


    legacy_document_id = (
        legacy_row[
            "candidate_document_id"
        ]
    )


    inside_core = (
        legacy_document_id
        in
        set(
            proposed_case[
                "evidence_document_ids"
            ]
        )
    )


    if inside_core:

        legacy_relationship = (
            "LEGACY_RECORD_INSIDE_EXERCISE_CORE"
        )

    else:

        legacy_relationship = (
            "LEGACY_RECORD_OUTSIDE_EXERCISE_CORE"
        )


    legacy_to_proposed_exercise_case_mapping.append(
        {
            "legacy_golden_case_id":
                legacy_row[
                    "golden_case_id"
                ],

            "legacy_candidate_document_id":
                legacy_document_id,

            "legacy_candidate_position":
                legacy_row[
                    "candidate_position"
                ],

            "proposed_exercise_unit_case_id":
                proposed_case[
                    "proposed_exercise_unit_case_id"
                ],

            "work_id":
                proposed_case[
                    "work_id"
                ],

            "edition_id":
                proposed_case[
                    "edition_id"
                ],

            "heading_path":
                proposed_case[
                    "heading_path"
                ],

            "relationship":
                legacy_relationship,

            "inside_resolved_exercise_core":
                inside_core,

            "supersession_action":
                "NOT_APPLIED",
        }
    )


# ============================================================
# 6) Mapping accounting
# ============================================================

legacy_mapping_relationship_counts = Counter(
    row[
        "relationship"
    ]
    for row
    in legacy_to_proposed_exercise_case_mapping
)


legacy_case_mapping_counts = Counter(
    row[
        "legacy_golden_case_id"
    ]
    for row
    in legacy_to_proposed_exercise_case_mapping
)


proposed_case_mapping_counts = Counter(
    row[
        "proposed_exercise_unit_case_id"
    ]
    for row
    in legacy_to_proposed_exercise_case_mapping
)


proposed_legacy_case_count_distribution = Counter(
    case[
        "legacy_record_level_case_count"
    ]
    for case
    in proposed_exercise_unit_cases
)


# ============================================================
# 7) Identify proposed units with no legacy selected record
# inside the true exercise core
# ============================================================

proposed_units_with_zero_legacy_inside_core = [
    case
    for case
    in proposed_exercise_unit_cases
    if (
        case[
            "legacy_inside_core_count"
        ]
        == 0
    )
]


proposed_units_with_legacy_outside_core = [
    case
    for case
    in proposed_exercise_unit_cases
    if (
        case[
            "legacy_outside_core_count"
        ]
        > 0
    )
]


# ============================================================
# 8) Independent evidence integrity validation
# ============================================================

proposed_case_evidence_failures = []

cross_unit_document_usage = defaultdict(
    list
)


for proposed_case in (
    proposed_exercise_unit_cases
):

    expected_heading_path = tuple(
        proposed_case[
            "heading_path"
        ]
    )


    for document_id in (
        proposed_case[
            "evidence_document_ids"
        ]
    ):

        if (
            document_id
            not in
            corpus_target_index
        ):

            proposed_case_evidence_failures.append(
                {
                    "proposed_case_id":
                        proposed_case[
                            "proposed_exercise_unit_case_id"
                        ],

                    "document_id":
                        document_id,

                    "reason":
                        "DOCUMENT_NOT_IN_ACCEPTED_CORPUS",
                }
            )

            continue


        record = (
            corpus_target_index[
                document_id
            ]
        )


        cross_unit_document_usage[
            document_id
        ].append(
            proposed_case[
                "proposed_exercise_unit_case_id"
            ]
        )


        if (
            record[
                "source_type"
            ]
            != "book"
        ):

            proposed_case_evidence_failures.append(
                {
                    "proposed_case_id":
                        proposed_case[
                            "proposed_exercise_unit_case_id"
                        ],

                    "document_id":
                        document_id,

                    "reason":
                        "NON_BOOK_EVIDENCE",
                }
            )


        locator = (
            record[
                "locator"
            ]
            or {}
        )

        metadata = (
            record[
                "source_metadata"
            ]
            or {}
        )


        record_heading_path = (
            locator.get(
                "heading_path"
            )
        )

        if (
            record_heading_path
            is None
        ):

            record_heading_path = (
                metadata.get(
                    "heading_path"
                )
            )


        if isinstance(
            record_heading_path,
            list,
        ):

            record_heading_key = tuple(
                str(
                    value
                )
                for value
                in record_heading_path
            )

        elif (
            record_heading_path
            is None
        ):

            record_heading_key = tuple()

        else:

            record_heading_key = (
                str(
                    record_heading_path
                ),
            )


        if (
            locator.get(
                "work_id"
            )
            !=
            proposed_case[
                "work_id"
            ]
        ):

            proposed_case_evidence_failures.append(
                {
                    "proposed_case_id":
                        proposed_case[
                            "proposed_exercise_unit_case_id"
                        ],

                    "document_id":
                        document_id,

                    "reason":
                        "WORK_ID_MISMATCH",
                }
            )


        if (
            locator.get(
                "edition_id"
            )
            !=
            proposed_case[
                "edition_id"
            ]
        ):

            proposed_case_evidence_failures.append(
                {
                    "proposed_case_id":
                        proposed_case[
                            "proposed_exercise_unit_case_id"
                        ],

                    "document_id":
                        document_id,

                    "reason":
                        "EDITION_ID_MISMATCH",
                }
            )


        if (
            record_heading_key
            !=
            expected_heading_path
        ):

            proposed_case_evidence_failures.append(
                {
                    "proposed_case_id":
                        proposed_case[
                            "proposed_exercise_unit_case_id"
                        ],

                    "document_id":
                        document_id,

                    "reason":
                        "HEADING_PATH_MISMATCH",
                }
            )


        if not (
            record[
                "text"
            ]
            or ""
        ).strip():

            proposed_case_evidence_failures.append(
                {
                    "proposed_case_id":
                        proposed_case[
                            "proposed_exercise_unit_case_id"
                        ],

                    "document_id":
                        document_id,

                    "reason":
                        "EMPTY_TEXT",
                }
            )


        if not (
            record[
                "locator"
            ]
        ):

            proposed_case_evidence_failures.append(
                {
                    "proposed_case_id":
                        proposed_case[
                            "proposed_exercise_unit_case_id"
                        ],

                    "document_id":
                        document_id,

                    "reason":
                        "MISSING_LOCATOR",
                }
            )


        if (
            record[
                "lineage"
            ]
            is None
        ):

            proposed_case_evidence_failures.append(
                {
                    "proposed_case_id":
                        proposed_case[
                            "proposed_exercise_unit_case_id"
                        ],

                    "document_id":
                        document_id,

                    "reason":
                        "MISSING_LINEAGE",
                }
            )


        if (
            record[
                "rights"
            ]
            is None
        ):

            proposed_case_evidence_failures.append(
                {
                    "proposed_case_id":
                        proposed_case[
                            "proposed_exercise_unit_case_id"
                        ],

                    "document_id":
                        document_id,

                    "reason":
                        "MISSING_RIGHTS",
                }
            )


        if (
            record[
                "attribution"
            ]
            is None
        ):

            proposed_case_evidence_failures.append(
                {
                    "proposed_case_id":
                        proposed_case[
                            "proposed_exercise_unit_case_id"
                        ],

                    "document_id":
                        document_id,

                    "reason":
                        "MISSING_ATTRIBUTION",
                }
            )


# ============================================================
# 9) Cross-unit evidence overlap audit
#
# Exercise units should not reuse the same accepted-corpus
# document as evidence.
# ============================================================

cross_unit_evidence_overlap = {
    document_id:
        proposed_case_ids

    for document_id, proposed_case_ids
    in cross_unit_document_usage.items()

    if (
        len(
            set(
                proposed_case_ids
            )
        )
        > 1
    )
}


# ============================================================
# 10) Independent stable-ID recomputation
# ============================================================

proposed_case_id_recomputation_failures = []


for proposed_case in (
    proposed_exercise_unit_cases
):

    recomputed_id = (
        build_proposed_exercise_unit_case_id(
            work_id=(
                proposed_case[
                    "work_id"
                ]
            ),
            edition_id=(
                proposed_case[
                    "edition_id"
                ]
            ),
            heading_path=(
                proposed_case[
                    "heading_path"
                ]
            ),
            exercise_core_document_ids=(
                proposed_case[
                    "evidence_document_ids"
                ]
            ),
        )
    )


    if (
        recomputed_id
        !=
        proposed_case[
            "proposed_exercise_unit_case_id"
        ]
    ):

        proposed_case_id_recomputation_failures.append(
            proposed_case[
                "proposed_exercise_unit_case_id"
            ]
        )


# ============================================================
# 11) Independently recompute evidence fingerprints
# ============================================================

proposed_evidence_fingerprint_failures = []


for proposed_case in (
    proposed_exercise_unit_cases
):

    recomputed_payload = []


    for document_id in (
        proposed_case[
            "evidence_document_ids"
        ]
    ):

        text = (
            corpus_target_index[
                document_id
            ][
                "text"
            ]
        )


        recomputed_payload.append(
            {
                "document_id":
                    document_id,

                "text_sha256":
                    hashlib.sha256(
                        text.encode(
                            "utf-8"
                        )
                    ).hexdigest(),
            }
        )


    recomputed_fingerprint = hashlib.sha256(
        stable_json_key(
            recomputed_payload
        ).encode(
            "utf-8"
        )
    ).hexdigest()


    if (
        recomputed_fingerprint
        !=
        proposed_case[
            "exercise_core_evidence_sha256"
        ]
    ):

        proposed_evidence_fingerprint_failures.append(
            proposed_case[
                "proposed_exercise_unit_case_id"
            ]
        )


# ============================================================
# 12) Proposed reconstruction registry
# ============================================================

EXERCISE_CASE_IDENTITY_RECONSTRUCTION_V1_DRAFT = {
    "version":
        "exercise_case_identity_reconstruction_v1_draft",

    "status":
        "PROPOSED_18_UNIT_CASES_VALIDATED_NOT_APPLIED",

    "legacy_record_level_case_count":
        len(
            legacy_to_proposed_exercise_case_mapping
        ),

    "proposed_exercise_unit_case_count":
        len(
            proposed_exercise_unit_cases
        ),

    "legacy_inside_core_count":
        legacy_mapping_relationship_counts[
            "LEGACY_RECORD_INSIDE_EXERCISE_CORE"
        ],

    "legacy_outside_core_count":
        legacy_mapping_relationship_counts[
            "LEGACY_RECORD_OUTSIDE_EXERCISE_CORE"
        ],

    "proposed_units_with_zero_legacy_inside_core":
        len(
            proposed_units_with_zero_legacy_inside_core
        ),

    "proposed_units_with_legacy_outside_core":
        len(
            proposed_units_with_legacy_outside_core
        ),

    "identity_reconstruction_status":
        "VALIDATED_PROPOSAL_NOT_APPLIED",

    "legacy_supersession_status":
        "NOT_STARTED",

    "global_golden_dataset_recount_status":
        "NOT_STARTED",

    "target_repair_status":
        "NOT_STARTED",

    "question_generation_status":
        "NOT_STARTED",

    "expected_answer_generation_status":
        "NOT_STARTED",

    "artifact_write_status":
        "NONE",
}


# ============================================================
# 13) Validation gates
# ============================================================

exact_18_proposed_cases_gate = (
    len(
        proposed_exercise_unit_cases
    )
    == 18
)


exact_38_legacy_mapping_rows_gate = (
    len(
        legacy_to_proposed_exercise_case_mapping
    )
    == 38
)


each_legacy_case_maps_once_gate = (
    len(
        legacy_case_mapping_counts
    )
    == 38

    and

    all(
        count == 1
        for count
        in legacy_case_mapping_counts.values()
    )
)


all_18_proposed_cases_receive_legacy_mapping_gate = (
    len(
        proposed_case_mapping_counts
    )
    == 18

    and

    all(
        count >= 1
        for count
        in proposed_case_mapping_counts.values()
    )
)


proposed_case_ids_unique_gate = (
    len(
        proposed_exercise_unit_cases_by_id
    )
    == 18
)


proposed_case_ids_recomputable_gate = (
    len(
        proposed_case_id_recomputation_failures
    )
    == 0
)


proposed_evidence_fingerprints_recomputable_gate = (
    len(
        proposed_evidence_fingerprint_failures
    )
    == 0
)


proposed_evidence_integrity_gate = (
    len(
        proposed_case_evidence_failures
    )
    == 0
)


no_cross_unit_evidence_overlap_gate = (
    len(
        cross_unit_evidence_overlap
    )
    == 0
)


legacy_28_inside_10_outside_gate = (
    legacy_mapping_relationship_counts[
        "LEGACY_RECORD_INSIDE_EXERCISE_CORE"
    ]
    == 28

    and

    legacy_mapping_relationship_counts[
        "LEGACY_RECORD_OUTSIDE_EXERCISE_CORE"
    ]
    == 10
)


proposal_not_applied_gate = all(
    case[
        "proposal_status"
    ]
    ==
    "PROPOSED_NOT_APPLIED"

    and

    case[
        "final_golden_membership"
    ]
    is False

    and

    case[
        "supersession_status"
    ]
    ==
    "NOT_APPLIED"

    for case
    in proposed_exercise_unit_cases
)


legacy_supersession_not_started_gate = all(
    row[
        "supersession_action"
    ]
    ==
    "NOT_APPLIED"

    for row
    in legacy_to_proposed_exercise_case_mapping
)


target_repair_not_started_gate = (
    EXERCISE_CASE_IDENTITY_RECONSTRUCTION_V1_DRAFT[
        "target_repair_status"
    ]
    ==
    "NOT_STARTED"
)


question_generation_not_started_gate = (
    EXERCISE_CASE_IDENTITY_RECONSTRUCTION_V1_DRAFT[
        "question_generation_status"
    ]
    ==
    "NOT_STARTED"
)


expected_answer_not_started_gate = (
    EXERCISE_CASE_IDENTITY_RECONSTRUCTION_V1_DRAFT[
        "expected_answer_generation_status"
    ]
    ==
    "NOT_STARTED"
)


no_artifact_write_gate = (
    EXERCISE_CASE_IDENTITY_RECONSTRUCTION_V1_DRAFT[
        "artifact_write_status"
    ]
    ==
    "NONE"
)


cell_38_gates = {
    "exact_18_proposed_exercise_unit_cases":
        exact_18_proposed_cases_gate,

    "exact_38_legacy_record_case_mapping_rows":
        exact_38_legacy_mapping_rows_gate,

    "every_legacy_case_maps_to_exactly_one_exercise_unit":
        each_legacy_case_maps_once_gate,

    "all_18_proposed_units_receive_at_least_one_legacy_mapping":
        all_18_proposed_cases_receive_legacy_mapping_gate,

    "proposed_exercise_unit_case_ids_unique":
        proposed_case_ids_unique_gate,

    "proposed_case_ids_independently_recomputable":
        proposed_case_ids_recomputable_gate,

    "exercise_core_evidence_fingerprints_recomputable":
        proposed_evidence_fingerprints_recomputable_gate,

    "all_proposed_case_evidence_preserves_structure_lineage_rights_attribution":
        proposed_evidence_integrity_gate,

    "no_document_shared_between_proposed_exercise_units":
        no_cross_unit_evidence_overlap_gate,

    "legacy_mapping_reproduces_28_inside_10_outside":
        legacy_28_inside_10_outside_gate,

    "proposed_case_replacement_not_applied":
        proposal_not_applied_gate,

    "legacy_case_supersession_not_started":
        legacy_supersession_not_started_gate,

    "target_repair_not_started":
        target_repair_not_started_gate,

    "question_generation_not_started":
        question_generation_not_started_gate,

    "expected_answer_generation_not_started":
        expected_answer_not_started_gate,

    "no_artifact_write":
        no_artifact_write_gate,
}


cell_38_pass = all(
    cell_38_gates.values()
)


# ============================================================
# 14) Output
# ============================================================

print(
    "=== EXERCISE-LEVEL CASE IDENTITY RECONSTRUCTION ==="
)

print(
    "Legacy record-level Golden Cases:",
    len(
        legacy_to_proposed_exercise_case_mapping
    )
)

print(
    "Proposed exercise-unit Cases:",
    len(
        proposed_exercise_unit_cases
    )
)


print(
    "\n=== LEGACY RELATIONSHIP COUNTS ==="
)

for relationship, count in sorted(
    legacy_mapping_relationship_counts.items()
):

    print(
        "-",
        relationship,
        ":",
        count,
    )


print(
    "\n=== LEGACY CASES PER PROPOSED UNIT ==="
)

for legacy_count, unit_count in sorted(
    proposed_legacy_case_count_distribution.items()
):

    print(
        "- legacy cases:",
        legacy_count,
        "| proposed units:",
        unit_count,
    )


print(
    "\nProposed units with ZERO previously selected record inside core:",
    len(
        proposed_units_with_zero_legacy_inside_core
    )
)

for case in (
    proposed_units_with_zero_legacy_inside_core
):

    print(
        "-",
        case[
            "heading_path"
        ],
        "| legacy cases:",
        case[
            "legacy_record_level_case_count"
        ],
        "| outside-core legacy cases:",
        case[
            "legacy_outside_core_count"
        ],
    )


print(
    "\nProposed units containing at least one previous outside-core candidate:",
    len(
        proposed_units_with_legacy_outside_core
    )
)

for case in (
    proposed_units_with_legacy_outside_core
):

    print(
        "-",
        case[
            "heading_path"
        ],
        "| inside:",
        case[
            "legacy_inside_core_count"
        ],
        "| outside:",
        case[
            "legacy_outside_core_count"
        ],
    )


print(
    "\n=== PROPOSED EXERCISE-UNIT CASES ==="
)

for index, case in enumerate(
    proposed_exercise_unit_cases,
    start=1,
):

    print(
        "\nUNIT",
        index,
        "/",
        len(
            proposed_exercise_unit_cases
        )
    )

    print(
        "Proposed ID:",
        case[
            "proposed_exercise_unit_case_id"
        ]
    )

    print(
        "Heading:",
        case[
            "heading_path"
        ]
    )

    print(
        "Boundary source:",
        case[
            "boundary_source"
        ]
    )

    print(
        "Evidence records:",
        case[
            "evidence_document_count"
        ]
    )

    print(
        "Evidence tokens:",
        case[
            "exercise_core_token_count"
        ]
    )

    print(
        "Legacy record-level cases:",
        case[
            "legacy_record_level_case_count"
        ]
    )

    print(
        "Legacy inside core:",
        case[
            "legacy_inside_core_count"
        ]
    )

    print(
        "Legacy outside core:",
        case[
            "legacy_outside_core_count"
        ]
    )

    print(
        "Evidence SHA256:",
        case[
            "exercise_core_evidence_sha256"
        ]
    )


print(
    "\n=== LEGACY -> PROPOSED MAPPING ==="
)

for row in sorted(
    legacy_to_proposed_exercise_case_mapping,
    key=lambda row: (
        row[
            "proposed_exercise_unit_case_id"
        ],
        row[
            "legacy_candidate_position"
        ],
    ),
):

    print(
        "-",
        row[
            "legacy_golden_case_id"
        ],
        "->",
        row[
            "proposed_exercise_unit_case_id"
        ],
        "| position:",
        row[
            "legacy_candidate_position"
        ],
        "| relationship:",
        row[
            "relationship"
        ],
    )


print(
    "\n=== RECONSTRUCTION INTEGRITY ==="
)

print(
    "Proposed evidence failures:",
    len(
        proposed_case_evidence_failures
    )
)

print(
    "Cross-unit evidence overlap:",
    len(
        cross_unit_evidence_overlap
    )
)

print(
    "Case-ID recomputation failures:",
    len(
        proposed_case_id_recomputation_failures
    )
)

print(
    "Evidence fingerprint failures:",
    len(
        proposed_evidence_fingerprint_failures
    )
)


print(
    "\n=== CODE CELL 38 GATES ==="
)

for name, passed in (
    cell_38_gates.items()
):

    print(
        name,
        ":",
        "PASS"
        if passed
        else "FAIL"
    )


print(
    "\nCODE CELL 38 GATE:",
    "PASS"
    if cell_38_pass
    else "FAIL"
)


assert cell_38_pass, (
    "Exercise-level Case Identity Reconstruction failed. "
    "Do not supersede the 38 record-level exercise cases until "
    "all 18 proposed exercise-unit identities and the complete "
    "38-to-18 traceability mapping are independently validated."
)
```

<li>چه Unitهای دیگری در Corpus وجود دارند که Sample قبلی نماینده‌ی آن‌ها نبوده است؛</li>

```python
# Code Cell 39

# ============================================================
# Full-Corpus Exercise Unit Universe Discovery — REPAIRED
#
# Root-cause repair:
# proposed_exercise_unit_cases created in Cell 38 do NOT store
# "structure_group_key" directly.
#
# Their structural identity is stored as:
#   work_id + edition_id + heading_path
#
# Therefore this Cell reconstructs the exact structural key
# instead of reading a nonexistent field.
#
# DISCOVERY ONLY.
#
# NO supersession.
# NO sampling rerun.
# NO final eligibility decision.
# NO boundary mutation.
# NO target generation.
# NO question generation.
# NO artifact writes.
# ============================================================


# ============================================================
# 1) Broad, explicit heading markers
#
# Evidence-derived from observed Book structures.
# These define a DISCOVERY universe only.
# ============================================================

EXERCISE_UNIT_DISCOVERY_HEADING_MARKERS = [
    "کاربرگ",
    "خودارزیابی",
    "خودآزمایی",
    "خود آزمایی",
    "تمرین",
    "پرسشنامه",
]


def normalize_heading_component(
    value,
):

    return (
        str(
            value
            or ""
        )
        .replace(
            "\u200c",
            " "
        )
        .strip()
    )


def exercise_heading_marker_hits(
    heading_path,
):

    hits = []


    for heading_component in (
        heading_path
        or []
    ):

        normalized_component = (
            normalize_heading_component(
                heading_component
            )
        )


        for marker in (
            EXERCISE_UNIT_DISCOVERY_HEADING_MARKERS
        ):

            normalized_marker = (
                normalize_heading_component(
                    marker
                )
            )


            if (
                normalized_marker
                in normalized_component
            ):

                hits.append(
                    {
                        "heading_component":
                            heading_component,

                        "marker":
                            marker,
                    }
                )


    return hits


# ============================================================
# 2) Canonical structural-key helper
#
# This must match the shape already used by
# book_heading_groups:
#
# (
#   work_id,
#   edition_id,
#   tuple(heading_path)
# )
#
# IMPORTANT:
# Do NOT normalize heading text here.
# Identity comparison must use the exact stored heading strings.
# ============================================================

def build_book_structure_group_key(
    work_id,
    edition_id,
    heading_path,
):

    return (
        work_id,
        edition_id,
        tuple(
            str(
                value
            )
            for value
            in (
                heading_path
                or []
            )
        ),
    )


# ============================================================
# 3) Inspect every unique Book Work / Edition / Heading Path
# ============================================================

full_book_structural_group_rows = []


for group_key, group_records in sorted(
    book_heading_groups.items(),
    key=lambda item: (
        str(
            item[
                0
            ][
                0
            ]
        ),
        str(
            item[
                0
            ][
                1
            ]
        ),
        item[
            0
        ][
            2
        ],
    ),
):

    work_id = (
        group_key[
            0
        ]
    )

    edition_id = (
        group_key[
            1
        ]
    )

    heading_path = list(
        group_key[
            2
        ]
    )


    marker_hits = (
        exercise_heading_marker_hits(
            heading_path
        )
    )


    reconstructed_group_key = (
        build_book_structure_group_key(
            work_id=work_id,
            edition_id=edition_id,
            heading_path=heading_path,
        )
    )


    full_book_structural_group_rows.append(
        {
            "work_id":
                work_id,

            "edition_id":
                edition_id,

            "heading_path":
                heading_path,

            "structure_group_key":
                group_key,

            "reconstructed_structure_group_key":
                reconstructed_group_key,

            "record_count":
                len(
                    group_records
                ),

            "token_count":
                sum(
                    record[
                        "token_count"
                    ]
                    for record
                    in group_records
                ),

            "table_record_count":
                sum(
                    1
                    for record
                    in group_records
                    if (
                        record[
                            "in_table"
                        ]
                        is True
                    )
                ),

            "heading_marker_hits":
                marker_hits,

            "exercise_like_heading":
                bool(
                    marker_hits
                ),
        }
    )


# ============================================================
# 4) Verify reconstructed keys reproduce book_heading_groups
# ============================================================

full_group_key_reconstruction_failures = [
    {
        "structure_group_key":
            row[
                "structure_group_key"
            ],

        "reconstructed_structure_group_key":
            row[
                "reconstructed_structure_group_key"
            ],
    }

    for row
    in full_book_structural_group_rows

    if (
        row[
            "structure_group_key"
        ]
        !=
        row[
            "reconstructed_structure_group_key"
        ]
    )
]


# ============================================================
# 5) Broad exercise-like structural universe
# ============================================================

full_exercise_like_group_rows = [
    row
    for row
    in full_book_structural_group_rows
    if (
        row[
            "exercise_like_heading"
        ]
        is True
    )
]


full_exercise_like_group_by_key = {
    row[
        "structure_group_key"
    ]:
        row

    for row
    in full_exercise_like_group_rows
}


# ============================================================
# 6) Reconstruct exact structural keys for the 18 proposed
# units from Cell 38
#
# REPAIR:
# proposed_exercise_unit_cases do not contain
# "structure_group_key".
# ============================================================

selected_sample_unit_key_rows = []


for case in (
    proposed_exercise_unit_cases
):

    reconstructed_key = (
        build_book_structure_group_key(
            work_id=(
                case[
                    "work_id"
                ]
            ),
            edition_id=(
                case[
                    "edition_id"
                ]
            ),
            heading_path=(
                case[
                    "heading_path"
                ]
            ),
        )
    )


    selected_sample_unit_key_rows.append(
        {
            "proposed_exercise_unit_case_id":
                case[
                    "proposed_exercise_unit_case_id"
                ],

            "work_id":
                case[
                    "work_id"
                ],

            "edition_id":
                case[
                    "edition_id"
                ],

            "heading_path":
                list(
                    case[
                        "heading_path"
                    ]
                ),

            "reconstructed_structure_group_key":
                reconstructed_key,
        }
    )


selected_sample_unit_keys = {
    row[
        "reconstructed_structure_group_key"
    ]
    for row
    in selected_sample_unit_key_rows
}


# ============================================================
# 7) Compare reconstructed 18-unit sample against full
# discovered universe
# ============================================================

full_discovered_unit_keys = set(
    full_exercise_like_group_by_key
)


selected_units_missing_from_full_discovery = sorted(
    selected_sample_unit_keys
    -
    full_discovered_unit_keys,
    key=str,
)


additional_full_corpus_units = sorted(
    full_discovered_unit_keys
    -
    selected_sample_unit_keys,
    key=str,
)


selected_sample_unit_coverage_count = (
    len(
        selected_sample_unit_keys
        &
        full_discovered_unit_keys
    )
)


# ============================================================
# 8) Build additional-unit inspection rows
#
# These are candidates only.
# No eligibility or boundary decision is made.
# ============================================================

additional_exercise_unit_candidate_rows = []


for group_key in (
    additional_full_corpus_units
):

    row = (
        full_exercise_like_group_by_key[
            group_key
        ]
    )

    group_records = (
        book_heading_groups[
            group_key
        ]
    )


    first_records_preview = [
        {
            "paragraph_number":
                record[
                    "paragraph_number"
                ],

            "document_id":
                record[
                    "document_id"
                ],

            "is_heading":
                record[
                    "is_heading"
                ],

            "in_table":
                record[
                    "in_table"
                ],

            "token_count":
                record[
                    "token_count"
                ],

            "text_preview":
                record[
                    "text"
                ][
                    :250
                ],
        }

        for record
        in group_records[
            :8
        ]
    ]


    additional_exercise_unit_candidate_rows.append(
        {
            "work_id":
                row[
                    "work_id"
                ],

            "edition_id":
                row[
                    "edition_id"
                ],

            "heading_path":
                row[
                    "heading_path"
                ],

            "structure_group_key":
                group_key,

            "record_count":
                row[
                    "record_count"
                ],

            "token_count":
                row[
                    "token_count"
                ],

            "table_record_count":
                row[
                    "table_record_count"
                ],

            "heading_marker_hits":
                row[
                    "heading_marker_hits"
                ],

            "first_records_preview":
                first_records_preview,

            "eligibility_status":
                "NOT_YET_EVALUATED",

            "boundary_status":
                "NOT_YET_EVALUATED",
        }
    )


# ============================================================
# 9) Marker distributions
# ============================================================

full_universe_marker_counts = Counter()


for row in (
    full_exercise_like_group_rows
):

    for hit in (
        row[
            "heading_marker_hits"
        ]
    ):

        full_universe_marker_counts[
            hit[
                "marker"
            ]
        ] += 1


additional_universe_marker_counts = Counter()


for row in (
    additional_exercise_unit_candidate_rows
):

    for hit in (
        row[
            "heading_marker_hits"
        ]
    ):

        additional_universe_marker_counts[
            hit[
                "marker"
            ]
        ] += 1


# ============================================================
# 10) Work / edition coverage
# ============================================================

full_exercise_units_by_work = Counter(
    row[
        "work_id"
    ]
    for row
    in full_exercise_like_group_rows
)


selected_sample_units_by_work = Counter(
    case[
        "work_id"
    ]
    for case
    in proposed_exercise_unit_cases
)


additional_units_by_work = Counter(
    row[
        "work_id"
    ]
    for row
    in additional_exercise_unit_candidate_rows
)


# ============================================================
# 11) Structural integrity of broad candidate universe
# ============================================================

full_exercise_universe_integrity_failures = []


for row in (
    full_exercise_like_group_rows
):

    group_records = (
        book_heading_groups[
            row[
                "structure_group_key"
            ]
        ]
    )


    if not (
        group_records
    ):

        full_exercise_universe_integrity_failures.append(
            {
                "heading_path":
                    row[
                        "heading_path"
                    ],

                "reason":
                    "EMPTY_STRUCTURAL_GROUP",
            }
        )

        continue


    if not (
        row[
            "heading_marker_hits"
        ]
    ):

        full_exercise_universe_integrity_failures.append(
            {
                "heading_path":
                    row[
                        "heading_path"
                    ],

                "reason":
                    "NO_DISCOVERY_MARKER",
            }
        )


    for record in (
        group_records
    ):

        document_id = (
            record[
                "document_id"
            ]
        )

        if (
            document_id
            not in
            corpus_target_index
        ):

            full_exercise_universe_integrity_failures.append(
                {
                    "heading_path":
                        row[
                            "heading_path"
                        ],

                    "document_id":
                        document_id,

                    "reason":
                        "DOCUMENT_NOT_IN_ACCEPTED_CORPUS",
                }
            )

            continue


        corpus_record = (
            corpus_target_index[
                document_id
            ]
        )


        if (
            corpus_record[
                "source_type"
            ]
            != "book"
        ):

            full_exercise_universe_integrity_failures.append(
                {
                    "heading_path":
                        row[
                            "heading_path"
                        ],

                    "document_id":
                        document_id,

                    "reason":
                        "NON_BOOK_RECORD",
                }
            )


        if not (
            corpus_record[
                "text"
            ]
            or ""
        ).strip():

            full_exercise_universe_integrity_failures.append(
                {
                    "heading_path":
                        row[
                            "heading_path"
                        ],

                    "document_id":
                        document_id,

                    "reason":
                        "EMPTY_TEXT",
                }
            )


        if not (
            corpus_record[
                "locator"
            ]
        ):

            full_exercise_universe_integrity_failures.append(
                {
                    "heading_path":
                        row[
                            "heading_path"
                        ],

                    "document_id":
                        document_id,

                    "reason":
                        "MISSING_LOCATOR",
                }
            )


        if (
            corpus_record[
                "lineage"
            ]
            is None
        ):

            full_exercise_universe_integrity_failures.append(
                {
                    "heading_path":
                        row[
                            "heading_path"
                        ],

                    "document_id":
                        document_id,

                    "reason":
                        "MISSING_LINEAGE",
                }
            )


# ============================================================
# 12) Validate reconstructed selected-unit keys
# ============================================================

selected_unit_key_uniqueness_failures = []


selected_unit_key_counts = Counter(
    row[
        "reconstructed_structure_group_key"
    ]
    for row
    in selected_sample_unit_key_rows
)


for key, count in (
    selected_unit_key_counts.items()
):

    if (
        count
        != 1
    ):

        selected_unit_key_uniqueness_failures.append(
            {
                "structure_group_key":
                    key,

                "count":
                    count,
            }
        )


# ============================================================
# 13) Determine whether 18-unit reconstruction is complete
#
# This is a DISCOVERY conclusion only.
# ============================================================

if (
    len(
        additional_exercise_unit_candidate_rows
    )
    == 0

    and

    len(
        selected_units_missing_from_full_discovery
    )
    == 0
):

    selected_sample_completeness_status = (
        "SELECTED_SAMPLE_COVERS_BROAD_DISCOVERED_UNIVERSE"
    )

else:

    selected_sample_completeness_status = (
        "SELECTED_SAMPLE_DOES_NOT_COVER_BROAD_DISCOVERED_UNIVERSE"
    )


sampling_rebuild_requirement = (
    "REQUIRED_BEFORE_EXERCISE_STRATUM_SUPERSESSION"
    if (
        len(
            additional_exercise_unit_candidate_rows
        )
        > 0
    )
    else
    "NOT_REQUIRED_BY_THIS_DISCOVERY"
)


# ============================================================
# 14) Discovery registry
# ============================================================

FULL_EXERCISE_UNIT_UNIVERSE_DISCOVERY_V1_DRAFT = {
    "version":
        "full_exercise_unit_universe_discovery_v1_draft",

    "status":
        "BROAD_HEADING_BASED_DISCOVERY_COMPLETE",

    "discovery_scope":
        "ALL_BOOK_STRUCTURAL_HEADING_GROUPS",

    "discovery_marker_policy":
        list(
            EXERCISE_UNIT_DISCOVERY_HEADING_MARKERS
        ),

    "broad_exercise_like_group_count":
        len(
            full_exercise_like_group_rows
        ),

    "selected_sample_reconstructed_unit_count":
        len(
            proposed_exercise_unit_cases
        ),

    "selected_sample_units_found_in_full_discovery":
        selected_sample_unit_coverage_count,

    "additional_full_corpus_unit_candidate_count":
        len(
            additional_exercise_unit_candidate_rows
        ),

    "selected_units_missing_from_full_discovery_count":
        len(
            selected_units_missing_from_full_discovery
        ),

    "selected_sample_completeness_status":
        selected_sample_completeness_status,

    "sampling_rebuild_requirement":
        sampling_rebuild_requirement,

    "additional_unit_eligibility_status":
        "NOT_YET_EVALUATED",

    "additional_unit_boundary_status":
        "NOT_YET_EVALUATED",

    "legacy_supersession_status":
        "NOT_STARTED",

    "target_repair_status":
        "NOT_STARTED",

    "question_generation_status":
        "NOT_STARTED",

    "expected_answer_generation_status":
        "NOT_STARTED",

    "artifact_write_status":
        "NONE",
}


# ============================================================
# 15) Validation gates
# ============================================================

full_group_key_reconstruction_gate = (
    len(
        full_group_key_reconstruction_failures
    )
    == 0
)


exact_18_selected_unit_keys_reconstructed_gate = (
    len(
        selected_sample_unit_key_rows
    )
    == 18

    and

    len(
        selected_sample_unit_keys
    )
    == 18
)


selected_unit_keys_unique_gate = (
    len(
        selected_unit_key_uniqueness_failures
    )
    == 0
)


all_18_reconstructed_units_present_in_discovery_gate = (
    len(
        selected_units_missing_from_full_discovery
    )
    == 0

    and

    selected_sample_unit_coverage_count
    ==
    18
)


broad_universe_nonempty_gate = (
    len(
        full_exercise_like_group_rows
    )
    >= 18
)


all_broad_groups_unique_gate = (
    len(
        full_discovered_unit_keys
    )
    ==
    len(
        full_exercise_like_group_rows
    )
)


broad_universe_integrity_gate = (
    len(
        full_exercise_universe_integrity_failures
    )
    == 0
)


additional_candidates_not_prejudged_gate = all(
    row[
        "eligibility_status"
    ]
    ==
    "NOT_YET_EVALUATED"

    and

    row[
        "boundary_status"
    ]
    ==
    "NOT_YET_EVALUATED"

    for row
    in additional_exercise_unit_candidate_rows
)


no_supersession_gate = (
    FULL_EXERCISE_UNIT_UNIVERSE_DISCOVERY_V1_DRAFT[
        "legacy_supersession_status"
    ]
    ==
    "NOT_STARTED"
)


no_target_repair_gate = (
    FULL_EXERCISE_UNIT_UNIVERSE_DISCOVERY_V1_DRAFT[
        "target_repair_status"
    ]
    ==
    "NOT_STARTED"
)


question_generation_not_started_gate = (
    FULL_EXERCISE_UNIT_UNIVERSE_DISCOVERY_V1_DRAFT[
        "question_generation_status"
    ]
    ==
    "NOT_STARTED"
)


expected_answer_not_started_gate = (
    FULL_EXERCISE_UNIT_UNIVERSE_DISCOVERY_V1_DRAFT[
        "expected_answer_generation_status"
    ]
    ==
    "NOT_STARTED"
)


no_artifact_write_gate = (
    FULL_EXERCISE_UNIT_UNIVERSE_DISCOVERY_V1_DRAFT[
        "artifact_write_status"
    ]
    ==
    "NONE"
)


cell_39_gates = {
    "all_book_structure_group_keys_exactly_reconstructable":
        full_group_key_reconstruction_gate,

    "exact_18_selected_exercise_unit_keys_reconstructed":
        exact_18_selected_unit_keys_reconstructed_gate,

    "all_18_selected_unit_structural_keys_unique":
        selected_unit_keys_unique_gate,

    "all_18_reconstructed_units_present_in_full_corpus_discovery":
        all_18_reconstructed_units_present_in_discovery_gate,

    "broad_exercise_like_universe_has_at_least_18_groups":
        broad_universe_nonempty_gate,

    "all_broad_discovered_structural_groups_unique":
        all_broad_groups_unique_gate,

    "all_broad_discovered_groups_preserve_book_text_locator_lineage":
        broad_universe_integrity_gate,

    "additional_full_corpus_units_not_semantically_prejudged":
        additional_candidates_not_prejudged_gate,

    "legacy_supersession_not_started":
        no_supersession_gate,

    "target_repair_not_started":
        no_target_repair_gate,

    "question_generation_not_started":
        question_generation_not_started_gate,

    "expected_answer_generation_not_started":
        expected_answer_not_started_gate,

    "no_artifact_write":
        no_artifact_write_gate,
}


cell_39_pass = all(
    cell_39_gates.values()
)


# ============================================================
# 16) Output
# ============================================================

print(
    "=== FULL-CORPUS EXERCISE UNIT UNIVERSE DISCOVERY ==="
)

print(
    "All Book structural groups:",
    len(
        full_book_structural_group_rows
    )
)

print(
    "Broad exercise-like structural groups:",
    len(
        full_exercise_like_group_rows
    )
)

print(
    "Previously reconstructed units:",
    len(
        proposed_exercise_unit_cases
    )
)

print(
    "Reconstructed structural keys:",
    len(
        selected_sample_unit_keys
    )
)

print(
    "Previously reconstructed units found in full discovery:",
    selected_sample_unit_coverage_count
)

print(
    "Additional full-corpus exercise-like group candidates:",
    len(
        additional_exercise_unit_candidate_rows
    )
)

print(
    "Previously reconstructed units missing from discovery:",
    len(
        selected_units_missing_from_full_discovery
    )
)


print(
    "\nSelected-sample completeness status:",
    selected_sample_completeness_status
)

print(
    "Sampling rebuild requirement:",
    sampling_rebuild_requirement
)


print(
    "\n=== STRUCTURAL-KEY VALIDATION ==="
)

print(
    "Full-group reconstruction failures:",
    len(
        full_group_key_reconstruction_failures
    )
)

print(
    "Selected-unit key uniqueness failures:",
    len(
        selected_unit_key_uniqueness_failures
    )
)


if (
    selected_units_missing_from_full_discovery
):

    print(
        "\nSelected reconstructed units missing from broad discovery:"
    )

    for key in (
        selected_units_missing_from_full_discovery
    ):

        print(
            "-",
            key
        )


print(
    "\n=== DISCOVERY MARKER COUNTS ==="
)

for marker, count in sorted(
    full_universe_marker_counts.items()
):

    print(
        "-",
        marker,
        ":",
        count,
    )


print(
    "\n=== FULL EXERCISE-LIKE UNIVERSE BY WORK ==="
)

for work_id, count in sorted(
    full_exercise_units_by_work.items()
):

    print(
        "-",
        work_id,
        ":",
        count,
    )


print(
    "\n=== PREVIOUSLY RECONSTRUCTED SAMPLE BY WORK ==="
)

for work_id, count in sorted(
    selected_sample_units_by_work.items()
):

    print(
        "-",
        work_id,
        ":",
        count,
    )


print(
    "\n=== ADDITIONAL FULL-CORPUS CANDIDATES BY WORK ==="
)

for work_id, count in sorted(
    additional_units_by_work.items()
):

    print(
        "-",
        work_id,
        ":",
        count,
    )


print(
    "\n=== ADDITIONAL DISCOVERY MARKER COUNTS ==="
)

for marker, count in sorted(
    additional_universe_marker_counts.items()
):

    print(
        "-",
        marker,
        ":",
        count,
    )


print(
    "\n=== ADDITIONAL EXERCISE-LIKE STRUCTURAL GROUPS ==="
)

if (
    additional_exercise_unit_candidate_rows
):

    for index, row in enumerate(
        additional_exercise_unit_candidate_rows,
        start=1,
    ):

        print(
            "\nADDITIONAL GROUP",
            index,
            "/",
            len(
                additional_exercise_unit_candidate_rows
            )
        )

        print(
            "Work:",
            row[
                "work_id"
            ]
        )

        print(
            "Edition:",
            row[
                "edition_id"
            ]
        )

        print(
            "Heading:",
            row[
                "heading_path"
            ]
        )

        print(
            "Marker hits:",
            row[
                "heading_marker_hits"
            ]
        )

        print(
            "Records:",
            row[
                "record_count"
            ],
            "| tokens:",
            row[
                "token_count"
            ],
            "| table records:",
            row[
                "table_record_count"
            ],
        )

        print(
            "First records:"
        )

        for preview in (
            row[
                "first_records_preview"
            ]
        ):

            print(
                "  - paragraph:",
                preview[
                    "paragraph_number"
                ],
                "| heading:",
                preview[
                    "is_heading"
                ],
                "| table:",
                preview[
                    "in_table"
                ],
                "| tokens:",
                preview[
                    "token_count"
                ],
                "| text:",
                preview[
                    "text_preview"
                ],
            )

else:

    print(
        "NONE"
    )


print(
    "\n=== DISCOVERY INTEGRITY FAILURES ==="
)

print(
    "Failures:",
    len(
        full_exercise_universe_integrity_failures
    )
)


if (
    full_exercise_universe_integrity_failures
):

    for failure in (
        full_exercise_universe_integrity_failures[
            :20
        ]
    ):

        print(
            "-",
            failure
        )


print(
    "\n=== CODE CELL 39 GATES ==="
)

for name, passed in (
    cell_39_gates.items()
):

    print(
        name,
        ":",
        "PASS"
        if passed
        else "FAIL"
    )


print(
    "\nCODE CELL 39 GATE:",
    "PASS"
    if cell_39_pass
    else "FAIL"
)


assert cell_39_pass, (
    "Full-corpus Exercise Unit discovery failed. "
    "Do not supersede the 38 record-level cases or rebuild "
    "sampling until the broader structural exercise universe "
    "is validly discovered."
)
```

<h3>Boundary Discovery for the 27 Additional Full-Corpus Exercise-Like Units</h3>

<p>
Cell 39 نشان داد Population وسیع Exercise-like در کل Book Corpus شامل
۴۵ Structural Unit است:
۱۸ Unit قبلاً از Sample قدیمی بازسازی شده بودند و
۲۷ Unit دیگر خارج از آن Sample قرار داشتند.
</p>

<p>
این ۲۷ Unit هنوز Final Golden Case نیستند.
در این مرحله فقط دو موضوع بررسی می‌شود:
</p>

<ul>
<li>
آیا Marker مربوط به Exercise/Self-assessment واقعاً در
<strong>آخرین جزء Heading Path</strong>
قرار دارد و بنابراین Unit Identity ساختاری معقول است؟
</li>

<li>
آیا داخل همان Heading Group یک Transition صریح مانند
<code>مطالعه بیشتر</code>
یا
<code>منابع بیشتر</code>
وجود دارد که Exercise Core را از Supplemental Material جدا کند؟
</li>
</ul>

<p>
برای Groupهایی که Marker انتقال صریح ندارند هیچ Boundary حدسی ساخته نمی‌شود.
آن‌ها با وضعیت
<code>NO_EXPLICIT_BOUNDARY_REQUIRES_FULL_INSPECTION</code>
باقی می‌مانند.
</p>

<p>
URL، Token Count، پایان Table و موقعیت Sample قدیمی
هیچ‌کدام Boundary محسوب نمی‌شوند.
</p>

<p>
هیچ Eligibility نهایی، Sampling، Supersession،
Target Evidence یا Question Generation در این Cell انجام نمی‌شود.
</p>

</div>

```python
# Code Cell 40

# ============================================================
# Additional Exercise-Like Unit Boundary Discovery
#
# Scope:
#   27 full-corpus structural groups discovered in Cell 39
#   but absent from the previous selected sample.
#
# DISCOVERY / INSPECTION ONLY.
#
# NO final eligibility.
# NO final boundary freeze.
# NO sampling rerun.
# NO supersession.
# NO target generation.
# NO question generation.
# NO artifact writes.
# ============================================================


# ============================================================
# 1) Validate that the exercise marker belongs to the terminal
# heading component of each additional structural group.
#
# This avoids accidentally treating a nested child group as an
# Exercise Unit merely because a parent heading contained a
# marker.
# ============================================================

additional_unit_structural_identity_rows = []


for candidate_row in (
    additional_exercise_unit_candidate_rows
):

    heading_path = list(
        candidate_row[
            "heading_path"
        ]
    )


    terminal_heading = (
        heading_path[
            -1
        ]
        if heading_path
        else ""
    )


    terminal_marker_hits = (
        exercise_heading_marker_hits(
            [
                terminal_heading
            ]
        )
    )


    any_heading_marker_hits = (
        candidate_row[
            "heading_marker_hits"
        ]
    )


    terminal_marker_valid = (
        len(
            terminal_marker_hits
        )
        > 0
    )


    additional_unit_structural_identity_rows.append(
        {
            "work_id":
                candidate_row[
                    "work_id"
                ],

            "edition_id":
                candidate_row[
                    "edition_id"
                ],

            "heading_path":
                heading_path,

            "structure_group_key":
                candidate_row[
                    "structure_group_key"
                ],

            "terminal_heading":
                terminal_heading,

            "all_heading_marker_hits":
                any_heading_marker_hits,

            "terminal_heading_marker_hits":
                terminal_marker_hits,

            "terminal_marker_valid":
                terminal_marker_valid,

            "structural_identity_status":
                (
                    "TERMINAL_EXERCISE_MARKER_CONFIRMED"
                    if terminal_marker_valid
                    else
                    "TERMINAL_EXERCISE_MARKER_NOT_CONFIRMED"
                ),
        }
    )


# ============================================================
# 2) Resolve exact group records and discover explicit
# supplemental transitions using the same narrow policy as
# Cell 35.
# ============================================================

additional_unit_boundary_candidate_rows = []


for identity_row in (
    additional_unit_structural_identity_rows
):

    group_key = (
        identity_row[
            "structure_group_key"
        ]
    )


    group_records = (
        book_heading_groups[
            group_key
        ]
    )


    explicit_candidates = []


    for index, record in enumerate(
        group_records
    ):

        marker_hits = (
            explicit_supplemental_marker_hits(
                record[
                    "text"
                ]
            )
        )


        if not marker_hits:
            continue


        explicit_candidates.append(
            {
                "index":
                    index,

                "paragraph_number":
                    record[
                        "paragraph_number"
                    ],

                "document_id":
                    record[
                        "document_id"
                    ],

                "record_class":
                    structural_record_class(
                        record
                    ),

                "marker_hits":
                    marker_hits,

                "text":
                    record[
                        "text"
                    ],
            }
        )


    # --------------------------------------------------------
    # Earliest explicit transition marker is only a
    # Boundary Candidate, not yet a frozen boundary.
    # --------------------------------------------------------

    if explicit_candidates:

        selected_boundary_candidate = sorted(
            explicit_candidates,
            key=lambda item: (
                item[
                    "index"
                ],
                item[
                    "document_id"
                ],
            ),
        )[0]


        supplemental_start_index = (
            selected_boundary_candidate[
                "index"
            ]
        )


        exercise_core_end_index = (
            supplemental_start_index
            - 1
        )


        if (
            exercise_core_end_index
            >= 0
        ):

            boundary_discovery_status = (
                "EXPLICIT_BOUNDARY_CANDIDATE_DISCOVERED"
            )

        else:

            boundary_discovery_status = (
                "INVALID_EXPLICIT_BOUNDARY_CANDIDATE"
            )


    else:

        selected_boundary_candidate = None

        supplemental_start_index = None

        exercise_core_end_index = None

        boundary_discovery_status = (
            "NO_EXPLICIT_BOUNDARY_REQUIRES_FULL_INSPECTION"
        )


    # --------------------------------------------------------
    # Candidate core — descriptive only.
    # --------------------------------------------------------

    if (
        exercise_core_end_index
        is not None

        and

        exercise_core_end_index
        >= 0
    ):

        candidate_core_records = (
            group_records[
                :exercise_core_end_index + 1
            ]
        )

    else:

        candidate_core_records = []


    structural_runs = (
        build_structural_runs(
            group_records
        )
    )


    full_sequence = []


    for index, record in enumerate(
        group_records
    ):

        full_sequence.append(
            {
                "index":
                    index,

                "paragraph_number":
                    record[
                        "paragraph_number"
                    ],

                "document_id":
                    record[
                        "document_id"
                    ],

                "record_class":
                    structural_record_class(
                        record
                    ),

                "is_heading":
                    record[
                        "is_heading"
                    ],

                "in_table":
                    record[
                        "in_table"
                    ],

                "url_like":
                    is_url_like_text(
                        record[
                            "text"
                        ]
                    ),

                "token_count":
                    record[
                        "token_count"
                    ],

                "exercise_cue_hits":
                    descriptive_cue_hits(
                        record[
                            "text"
                        ]
                    ),

                "explicit_supplemental_marker_hits":
                    explicit_supplemental_marker_hits(
                        record[
                            "text"
                        ]
                    ),

                "text":
                    record[
                        "text"
                    ],
            }
        )


    additional_unit_boundary_candidate_rows.append(
        {
            "work_id":
                identity_row[
                    "work_id"
                ],

            "edition_id":
                identity_row[
                    "edition_id"
                ],

            "heading_path":
                identity_row[
                    "heading_path"
                ],

            "structure_group_key":
                group_key,

            "terminal_marker_valid":
                identity_row[
                    "terminal_marker_valid"
                ],

            "structural_identity_status":
                identity_row[
                    "structural_identity_status"
                ],

            "group_record_count":
                len(
                    group_records
                ),

            "group_total_tokens":
                sum(
                    record[
                        "token_count"
                    ]
                    for record
                    in group_records
                ),

            "table_record_count":
                sum(
                    1
                    for record
                    in group_records
                    if (
                        record[
                            "in_table"
                        ]
                        is True
                    )
                ),

            "url_like_record_count":
                sum(
                    1
                    for record
                    in group_records
                    if is_url_like_text(
                        record[
                            "text"
                        ]
                    )
                ),

            "explicit_boundary_candidates":
                explicit_candidates,

            "selected_explicit_boundary_candidate":
                selected_boundary_candidate,

            "exercise_core_start_index":
                (
                    0
                    if candidate_core_records
                    else None
                ),

            "exercise_core_end_index":
                exercise_core_end_index,

            "supplemental_start_index":
                supplemental_start_index,

            "candidate_core_record_count":
                len(
                    candidate_core_records
                ),

            "candidate_core_token_count":
                sum(
                    record[
                        "token_count"
                    ]
                    for record
                    in candidate_core_records
                ),

            "candidate_core_document_ids":
                [
                    record[
                        "document_id"
                    ]
                    for record
                    in candidate_core_records
                ],

            "structural_runs":
                structural_runs,

            "full_sequence":
                full_sequence,

            "boundary_discovery_status":
                boundary_discovery_status,

            "eligibility_status":
                "NOT_YET_EVALUATED",

            "final_boundary_status":
                "NOT_FROZEN",
        }
    )


# ============================================================
# 3) Aggregate statuses
# ============================================================

additional_structural_identity_status_counts = Counter(
    row[
        "structural_identity_status"
    ]
    for row
    in additional_unit_boundary_candidate_rows
)


additional_boundary_status_counts = Counter(
    row[
        "boundary_discovery_status"
    ]
    for row
    in additional_unit_boundary_candidate_rows
)


additional_explicit_boundary_rows = [
    row
    for row
    in additional_unit_boundary_candidate_rows
    if (
        row[
            "boundary_discovery_status"
        ]
        ==
        "EXPLICIT_BOUNDARY_CANDIDATE_DISCOVERED"
    )
]


additional_unresolved_boundary_rows = [
    row
    for row
    in additional_unit_boundary_candidate_rows
    if (
        row[
            "boundary_discovery_status"
        ]
        ==
        "NO_EXPLICIT_BOUNDARY_REQUIRES_FULL_INSPECTION"
    )
]


additional_invalid_boundary_rows = [
    row
    for row
    in additional_unit_boundary_candidate_rows
    if (
        row[
            "boundary_discovery_status"
        ]
        ==
        "INVALID_EXPLICIT_BOUNDARY_CANDIDATE"
    )
]


additional_terminal_marker_failures = [
    row
    for row
    in additional_unit_boundary_candidate_rows
    if (
        row[
            "terminal_marker_valid"
        ]
        is False
    )
]


# ============================================================
# 4) Independently validate every explicit candidate core
# is an exact contiguous prefix of the structural group.
# ============================================================

additional_candidate_core_integrity_failures = []


for row in (
    additional_explicit_boundary_rows
):

    group_records = (
        book_heading_groups[
            row[
                "structure_group_key"
            ]
        ]
    )


    expected_core_ids = [
        record[
            "document_id"
        ]
        for record
        in group_records[
            :row[
                "exercise_core_end_index"
            ] + 1
        ]
    ]


    if (
        expected_core_ids
        !=
        row[
            "candidate_core_document_ids"
        ]
    ):

        additional_candidate_core_integrity_failures.append(
            {
                "heading_path":
                    row[
                        "heading_path"
                    ],

                "reason":
                    "NONCONTIGUOUS_CANDIDATE_CORE",
            }
        )


    if (
        row[
            "candidate_core_record_count"
        ]
        <= 0
    ):

        additional_candidate_core_integrity_failures.append(
            {
                "heading_path":
                    row[
                        "heading_path"
                    ],

                "reason":
                    "EMPTY_CANDIDATE_CORE",
            }
        )


    if (
        group_records[
            0
        ][
            "is_heading"
        ]
        is not True
    ):

        additional_candidate_core_integrity_failures.append(
            {
                "heading_path":
                    row[
                        "heading_path"
                    ],

                "reason":
                    "GROUP_DOES_NOT_BEGIN_WITH_HEADING",
            }
        )


# ============================================================
# 5) Corpus / lineage integrity
# ============================================================

additional_unit_record_integrity_failures = []


for row in (
    additional_unit_boundary_candidate_rows
):

    for item in (
        row[
            "full_sequence"
        ]
    ):

        document_id = (
            item[
                "document_id"
            ]
        )


        if (
            document_id
            not in
            corpus_target_index
        ):

            additional_unit_record_integrity_failures.append(
                {
                    "document_id":
                        document_id,

                    "reason":
                        "DOCUMENT_NOT_IN_ACCEPTED_CORPUS",
                }
            )

            continue


        record = (
            corpus_target_index[
                document_id
            ]
        )


        if (
            record[
                "source_type"
            ]
            != "book"
        ):

            additional_unit_record_integrity_failures.append(
                {
                    "document_id":
                        document_id,

                    "reason":
                        "NON_BOOK_RECORD",
                }
            )


        if not (
            record[
                "text"
            ]
            or ""
        ).strip():

            additional_unit_record_integrity_failures.append(
                {
                    "document_id":
                        document_id,

                    "reason":
                        "EMPTY_TEXT",
                }
            )


        if not (
            record[
                "locator"
            ]
        ):

            additional_unit_record_integrity_failures.append(
                {
                    "document_id":
                        document_id,

                    "reason":
                        "MISSING_LOCATOR",
                }
            )


        if (
            record[
                "lineage"
            ]
            is None
        ):

            additional_unit_record_integrity_failures.append(
                {
                    "document_id":
                        document_id,

                    "reason":
                        "MISSING_LINEAGE",
                }
            )


        if (
            record[
                "rights"
            ]
            is None
        ):

            additional_unit_record_integrity_failures.append(
                {
                    "document_id":
                        document_id,

                    "reason":
                        "MISSING_RIGHTS",
                }
            )


        if (
            record[
                "attribution"
            ]
            is None
        ):

            additional_unit_record_integrity_failures.append(
                {
                    "document_id":
                        document_id,

                    "reason":
                        "MISSING_ATTRIBUTION",
                }
            )


# ============================================================
# 6) Descriptive distributions
# ============================================================

additional_group_record_counts = [
    row[
        "group_record_count"
    ]
    for row
    in additional_unit_boundary_candidate_rows
]


additional_group_token_counts = [
    row[
        "group_total_tokens"
    ]
    for row
    in additional_unit_boundary_candidate_rows
]


additional_explicit_core_record_counts = [
    row[
        "candidate_core_record_count"
    ]
    for row
    in additional_explicit_boundary_rows
]


additional_explicit_core_token_counts = [
    row[
        "candidate_core_token_count"
    ]
    for row
    in additional_explicit_boundary_rows
]


# ============================================================
# 7) Registry
# ============================================================

ADDITIONAL_EXERCISE_UNIT_BOUNDARY_DISCOVERY_V1_DRAFT = {
    "version":
        "additional_exercise_unit_boundary_discovery_v1_draft",

    "status":
        "ADDITIONAL_27_STRUCTURALLY_INSPECTED_NOT_FROZEN",

    "additional_unit_count":
        len(
            additional_unit_boundary_candidate_rows
        ),

    "terminal_marker_confirmed_count":
        sum(
            1
            for row
            in additional_unit_boundary_candidate_rows
            if (
                row[
                    "terminal_marker_valid"
                ]
                is True
            )
        ),

    "explicit_boundary_candidate_count":
        len(
            additional_explicit_boundary_rows
        ),

    "requires_full_inspection_count":
        len(
            additional_unresolved_boundary_rows
        ),

    "invalid_boundary_candidate_count":
        len(
            additional_invalid_boundary_rows
        ),

    "eligibility_status":
        "NOT_YET_EVALUATED",

    "final_boundary_status":
        "NOT_FROZEN",

    "sampling_rebuild_status":
        "NOT_STARTED",

    "legacy_supersession_status":
        "NOT_STARTED",

    "target_repair_status":
        "NOT_STARTED",

    "question_generation_status":
        "NOT_STARTED",

    "expected_answer_generation_status":
        "NOT_STARTED",

    "artifact_write_status":
        "NONE",
}


# ============================================================
# 8) Validation gates
# ============================================================

exact_27_additional_units_gate = (
    len(
        additional_unit_boundary_candidate_rows
    )
    == 27
)


all_terminal_markers_confirmed_gate = (
    len(
        additional_terminal_marker_failures
    )
    == 0
)


all_27_boundary_statuses_accounted_gate = (
    sum(
        additional_boundary_status_counts.values()
    )
    == 27
)


no_invalid_explicit_boundary_gate = (
    len(
        additional_invalid_boundary_rows
    )
    == 0
)


explicit_candidate_core_integrity_gate = (
    len(
        additional_candidate_core_integrity_failures
    )
    == 0
)


all_additional_record_integrity_gate = (
    len(
        additional_unit_record_integrity_failures
    )
    == 0
)


eligibility_not_prejudged_gate = all(
    row[
        "eligibility_status"
    ]
    ==
    "NOT_YET_EVALUATED"

    for row
    in additional_unit_boundary_candidate_rows
)


final_boundary_not_frozen_gate = all(
    row[
        "final_boundary_status"
    ]
    ==
    "NOT_FROZEN"

    for row
    in additional_unit_boundary_candidate_rows
)


sampling_not_started_gate = (
    ADDITIONAL_EXERCISE_UNIT_BOUNDARY_DISCOVERY_V1_DRAFT[
        "sampling_rebuild_status"
    ]
    ==
    "NOT_STARTED"
)


supersession_not_started_gate = (
    ADDITIONAL_EXERCISE_UNIT_BOUNDARY_DISCOVERY_V1_DRAFT[
        "legacy_supersession_status"
    ]
    ==
    "NOT_STARTED"
)


target_repair_not_started_gate = (
    ADDITIONAL_EXERCISE_UNIT_BOUNDARY_DISCOVERY_V1_DRAFT[
        "target_repair_status"
    ]
    ==
    "NOT_STARTED"
)


question_generation_not_started_gate = (
    ADDITIONAL_EXERCISE_UNIT_BOUNDARY_DISCOVERY_V1_DRAFT[
        "question_generation_status"
    ]
    ==
    "NOT_STARTED"
)


expected_answer_not_started_gate = (
    ADDITIONAL_EXERCISE_UNIT_BOUNDARY_DISCOVERY_V1_DRAFT[
        "expected_answer_generation_status"
    ]
    ==
    "NOT_STARTED"
)


no_artifact_write_gate = (
    ADDITIONAL_EXERCISE_UNIT_BOUNDARY_DISCOVERY_V1_DRAFT[
        "artifact_write_status"
    ]
    ==
    "NONE"
)


cell_40_gates = {
    "exact_27_additional_exercise_like_units_inspected":
        exact_27_additional_units_gate,

    "exercise_marker_confirmed_in_terminal_heading_for_all_27":
        all_terminal_markers_confirmed_gate,

    "all_27_receive_boundary_discovery_status":
        all_27_boundary_statuses_accounted_gate,

    "no_invalid_explicit_boundary_candidate":
        no_invalid_explicit_boundary_gate,

    "all_explicit_candidate_cores_contiguous_and_valid":
        explicit_candidate_core_integrity_gate,

    "all_additional_unit_records_preserve_text_locator_lineage_rights_attribution":
        all_additional_record_integrity_gate,

    "additional_unit_eligibility_not_prejudged":
        eligibility_not_prejudged_gate,

    "final_boundary_not_frozen":
        final_boundary_not_frozen_gate,

    "sampling_rebuild_not_started":
        sampling_not_started_gate,

    "legacy_supersession_not_started":
        supersession_not_started_gate,

    "target_repair_not_started":
        target_repair_not_started_gate,

    "question_generation_not_started":
        question_generation_not_started_gate,

    "expected_answer_generation_not_started":
        expected_answer_not_started_gate,

    "no_artifact_write":
        no_artifact_write_gate,
}


cell_40_pass = all(
    cell_40_gates.values()
)


# ============================================================
# 9) Output
# ============================================================

print(
    "=== ADDITIONAL 27 EXERCISE-LIKE UNIT BOUNDARY DISCOVERY ==="
)

print(
    "Additional units inspected:",
    len(
        additional_unit_boundary_candidate_rows
    )
)

print(
    "Terminal heading marker confirmed:",
    sum(
        1
        for row
        in additional_unit_boundary_candidate_rows
        if (
            row[
                "terminal_marker_valid"
            ]
            is True
        )
    )
)

print(
    "Explicit boundary candidates:",
    len(
        additional_explicit_boundary_rows
    )
)

print(
    "Require full inspection:",
    len(
        additional_unresolved_boundary_rows
    )
)

print(
    "Invalid boundary candidates:",
    len(
        additional_invalid_boundary_rows
    )
)


print(
    "\n=== STRUCTURAL IDENTITY STATUS ==="
)

for status, count in sorted(
    additional_structural_identity_status_counts.items()
):

    print(
        "-",
        status,
        ":",
        count,
    )


print(
    "\n=== BOUNDARY DISCOVERY STATUS ==="
)

for status, count in sorted(
    additional_boundary_status_counts.items()
):

    print(
        "-",
        status,
        ":",
        count,
    )


print(
    "\n=== ADDITIONAL GROUP DISTRIBUTIONS ==="
)

print(
    "Group record count:",
    {
        "min":
            min(
                additional_group_record_counts
            ),

        "median":
            empirical_nearest_rank(
                additional_group_record_counts,
                0.50,
            ),

        "max":
            max(
                additional_group_record_counts
            ),
    }
)

print(
    "Group token count:",
    {
        "min":
            min(
                additional_group_token_counts
            ),

        "median":
            empirical_nearest_rank(
                additional_group_token_counts,
                0.50,
            ),

        "max":
            max(
                additional_group_token_counts
            ),
    }
)


if (
    additional_explicit_boundary_rows
):

    print(
        "Explicit candidate core record count:",
        {
            "min":
                min(
                    additional_explicit_core_record_counts
                ),

            "median":
                empirical_nearest_rank(
                    additional_explicit_core_record_counts,
                    0.50,
                ),

            "max":
                max(
                    additional_explicit_core_record_counts
                ),
        }
    )

    print(
        "Explicit candidate core token count:",
        {
            "min":
                min(
                    additional_explicit_core_token_counts
                ),

            "median":
                empirical_nearest_rank(
                    additional_explicit_core_token_counts,
                    0.50,
                ),

            "max":
                max(
                    additional_explicit_core_token_counts
                ),
        }
    )


print(
    "\n=== EXPLICIT BOUNDARY CANDIDATES ==="
)

for row in (
    additional_explicit_boundary_rows
):

    print(
        "\nHeading:",
        row[
            "heading_path"
        ]
    )

    print(
        "Boundary:",
        json.dumps(
            row[
                "selected_explicit_boundary_candidate"
            ],
            ensure_ascii=False,
            sort_keys=True,
        )
    )

    print(
        "Candidate core:",
        {
            "records":
                row[
                    "candidate_core_record_count"
                ],

            "tokens":
                row[
                    "candidate_core_token_count"
                ],

            "end_index":
                row[
                    "exercise_core_end_index"
                ],
        }
    )


print(
    "\n=== UNRESOLVED ADDITIONAL UNITS REQUIRING FULL INSPECTION ==="
)

for row in (
    additional_unresolved_boundary_rows
):

    print(
        "\nHeading:",
        row[
            "heading_path"
        ]
    )

    print(
        "Records:",
        row[
            "group_record_count"
        ],
        "| tokens:",
        row[
            "group_total_tokens"
        ],
        "| table records:",
        row[
            "table_record_count"
        ],
        "| URL-like:",
        row[
            "url_like_record_count"
        ],
    )

    print(
        "Structural runs:",
        tuple(
            run[
                "record_class"
            ]
            for run
            in row[
                "structural_runs"
            ]
        )
    )

    print(
        "Full sequence:"
    )

    for item in (
        row[
            "full_sequence"
        ]
    ):

        print(
            "- index:",
            item[
                "index"
            ],
            "| paragraph:",
            item[
                "paragraph_number"
            ],
            "| class:",
            item[
                "record_class"
            ],
            "| URL:",
            item[
                "url_like"
            ],
            "| cues:",
            item[
                "exercise_cue_hits"
            ],
            "| tokens:",
            item[
                "token_count"
            ],
        )

        print(
            "  TEXT:",
            item[
                "text"
            ]
        )


print(
    "\n=== INTEGRITY FAILURES ==="
)

print(
    "Terminal-marker failures:",
    len(
        additional_terminal_marker_failures
    )
)

print(
    "Candidate-core integrity failures:",
    len(
        additional_candidate_core_integrity_failures
    )
)

print(
    "Record integrity failures:",
    len(
        additional_unit_record_integrity_failures
    )
)


print(
    "\n=== CODE CELL 40 GATES ==="
)

for name, passed in (
    cell_40_gates.items()
):

    print(
        name,
        ":",
        "PASS"
        if passed
        else "FAIL"
    )


print(
    "\nCODE CELL 40 GATE:",
    "PASS"
    if cell_40_pass
    else "FAIL"
)


assert cell_40_pass, (
    "Additional Exercise Unit boundary discovery failed. "
    "Do not rebuild the exercise-unit population or rerun "
    "sampling until all 27 additional structural groups have "
    "valid terminal exercise markers and their explicit versus "
    "unresolved boundaries are fully accounted for."
)
```

```python
# Code Cell 41

# ============================================================
# Adjudicate 17 Additional Unresolved Exercise Boundaries
#
# Then combine:
#   18 previously resolved units
# + 10 new explicit-boundary units
# + 17 newly adjudicated units
# = 45 broad exercise-like structural units
#
# Boundary only.
#
# NO final eligibility.
# NO sampling rerun.
# NO case supersession.
# NO target repair.
# NO question generation.
# NO artifact writes.
# ============================================================


# ============================================================
# 1) Explicit adjudication for the 17 unresolved additional
# units exposed in Cell 40
# ============================================================

ADDITIONAL_UNRESOLVED_BOUNDARY_ADJUDICATION = {

    (
        "book:adult-financial-literacy-course",
        "book:adult-financial-literacy-course:fa-source",
        (
            "درس 14: یافتن شغل",
            "خودارزیابی 9: کار داوطلبانه در نگاه من",
        ),
    ): {
        "decision":
            "FULL_GROUP_IS_EXERCISE_UNIT",

        "exercise_core_start_index":
            0,

        "exercise_core_end_index":
            1,

        "first_nonexercise_index":
            None,

        "basis":
            (
                "The heading and sole following narrative form one "
                "self-assessment question about volunteer work."
            ),
    },


    (
        "book:adult-financial-literacy-course",
        "book:adult-financial-literacy-course:fa-source",
        (
            "درس 19: چگونگی سرمایه‌گذاری",
            "خودارزیابی 1۲: اولین سرمایه‌گذاری من",
        ),
    ): {
        "decision":
            "FULL_GROUP_IS_EXERCISE_UNIT",

        "exercise_core_start_index":
            0,

        "exercise_core_end_index":
            2,

        "first_nonexercise_index":
            None,

        "basis":
            (
                "Both narrative records are continuous self-assessment "
                "questions about the learner's first investment."
            ),
    },


    (
        "book:adult-financial-literacy-course",
        "book:adult-financial-literacy-course:fa-source",
        (
            "درس 23: مدیریت ریسک",
            "خودارزیابی ۱۶: مواجهه من با ریسک",
        ),
    ): {
        "decision":
            "FULL_GROUP_IS_EXERCISE_UNIT",

        "exercise_core_start_index":
            0,

        "exercise_core_end_index":
            1,

        "first_nonexercise_index":
            None,

        "basis":
            (
                "The only narrative record is the self-assessment prompt "
                "about the learner's preferred risk-management approach."
            ),
    },


    (
        "book:adult-financial-literacy-course",
        "book:adult-financial-literacy-course:fa-source",
        (
            "درس ۲: پلکان توانمندی مالی",
            "کاربرگ 1: نامه‌ای به خودتان",
        ),
    ): {
        "decision":
            "FULL_GROUP_IS_EXERCISE_UNIT",

        "exercise_core_start_index":
            0,

        "exercise_core_end_index":
            2,

        "first_nonexercise_index":
            None,

        "basis":
            (
                "Both following narrative records are worksheet "
                "instructions for constructing a future financial vision "
                "and writing a letter to one's future self."
            ),
    },


    (
        "book:adult-financial-literacy-course",
        "book:adult-financial-literacy-course:fa-source",
        (
            "درس ۲۵: مؤلفه‌های مالی در تصمیم‌گیری مالی",
            "خودارزیابی ۲۰: سبک پاسخگویی به تمایلات",
        ),
    ): {
        "decision":
            "INTERNAL_CONTENT_TRANSITION_BOUNDARY",

        "exercise_core_start_index":
            0,

        "exercise_core_end_index":
            1,

        "first_nonexercise_index":
            2,

        "basis":
            (
                "Index 1 contains the self-assessment task. "
                "Index 2 explicitly starts a new conceptual section: "
                "'۳. شاخص سوم تصمیم‌گیری مالی:هزینه فرصت.'."
            ),
    },


    (
        "book:trainer-training-financial-literacy",
        "book:trainer-training-financial-literacy:fa-source",
        (
            "بودجه‌بندی",
            "تمرین 8",
        ),
    ): {
        "decision":
            "FULL_GROUP_IS_EXERCISE_UNIT",

        "exercise_core_start_index":
            0,

        "exercise_core_end_index":
            1,

        "first_nonexercise_index":
            None,

        "basis":
            (
                "The sole narrative record is the complete exercise "
                "instruction for interviewing participants about expenses."
            ),
    },


    (
        "book:trainer-training-financial-literacy",
        "book:trainer-training-financial-literacy:fa-source",
        (
            "خرید هوشمندانه",
            "تمرین 7",
        ),
    ): {
        "decision":
            "FULL_GROUP_IS_EXERCISE_UNIT",

        "exercise_core_start_index":
            0,

        "exercise_core_end_index":
            4,

        "first_nonexercise_index":
            None,

        "basis":
            (
                "All four narrative records are continuous exercise "
                "prompts about past purchases and smart-buying behavior."
            ),
    },


    (
        "book:trainer-training-financial-literacy",
        "book:trainer-training-financial-literacy:fa-source",
        (
            "صندوق اضطراری",
            "خودارزیابی 4",
        ),
    ): {
        "decision":
            "FULL_GROUP_IS_EXERCISE_UNIT",

        "exercise_core_start_index":
            0,

        "exercise_core_end_index":
            2,

        "first_nonexercise_index":
            None,

        "basis":
            (
                "Both narrative records form one self-assessment about "
                "past and future emergency-fund needs."
            ),
    },


    (
        "book:trainer-training-financial-literacy",
        "book:trainer-training-financial-literacy:fa-source",
        (
            "مبادلۀ کالابه‌کالا",
            "تمرین 1",
        ),
    ): {
        "decision":
            "FULL_GROUP_IS_EXERCISE_UNIT",

        "exercise_core_start_index":
            0,

        "exercise_core_end_index":
            6,

        "first_nonexercise_index":
            None,

        "basis":
            (
                "All narrative records are successive parts of the same "
                "exercise about nominal/real income, prices and inflation."
            ),
    },


    (
        "book:trainer-training-financial-literacy",
        "book:trainer-training-financial-literacy:fa-source",
        (
            "مبادلۀ کالابه‌کالا",
            "تمرین 2",
        ),
    ): {
        "decision":
            "FULL_GROUP_IS_EXERCISE_UNIT",

        "exercise_core_start_index":
            0,

        "exercise_core_end_index":
            1,

        "first_nonexercise_index":
            None,

        "basis":
            (
                "The sole narrative record is the complete exercise "
                "about exchange-rate supply, demand and intervention."
            ),
    },


    (
        "book:trainer-training-financial-literacy",
        "book:trainer-training-financial-literacy:fa-source",
        (
            "هدف پس‌اندازی",
            "تمرین 5",
        ),
    ): {
        "decision":
            "FULL_GROUP_IS_EXERCISE_UNIT",

        "exercise_core_start_index":
            0,

        "exercise_core_end_index":
            2,

        "first_nonexercise_index":
            None,

        "basis":
            (
                "Both narrative records are related exercise tasks "
                "for designing saving goals for self and age groups."
            ),
    },


    (
        "book:trainer-training-financial-literacy",
        "book:trainer-training-financial-literacy:fa-source",
        (
            "هدف‌گذاری هوشمندانه",
            "تمرین 6",
        ),
    ): {
        "decision":
            "FULL_GROUP_IS_EXERCISE_UNIT",

        "exercise_core_start_index":
            0,

        "exercise_core_end_index":
            2,

        "first_nonexercise_index":
            None,

        "basis":
            (
                "Both narrative records are exercise instructions about "
                "future vision, SMART formulation and personal mission."
            ),
    },


    (
        "book:trainer-training-financial-literacy",
        "book:trainer-training-financial-literacy:fa-source",
        (
            "پس‌انداز",
            "تمرین 3",
        ),
    ): {
        "decision":
            "FULL_GROUP_IS_EXERCISE_UNIT",

        "exercise_core_start_index":
            0,

        "exercise_core_end_index":
            1,

        "first_nonexercise_index":
            None,

        "basis":
            (
                "The sole narrative record is the full exercise on "
                "present-versus-future preference and financial balance."
            ),
    },


    (
        "book:trainer-training-financial-literacy",
        "book:trainer-training-financial-literacy:fa-source",
        (
            "پس‌انداز",
            "تمرین 4",
        ),
    ): {
        "decision":
            "FULL_GROUP_IS_EXERCISE_UNIT",

        "exercise_core_start_index":
            0,

        "exercise_core_end_index":
            1,

        "first_nonexercise_index":
            None,

        "basis":
            (
                "The sole narrative record is the complete exercise "
                "about current pleasure, future benefit and decisions."
            ),
    },


    (
        "book:trainer-training-financial-literacy",
        "book:trainer-training-financial-literacy:fa-source",
        (
            "پول‌توجیبی",
            "تمرین 9",
        ),
    ): {
        "decision":
            "FULL_GROUP_IS_EXERCISE_UNIT",

        "exercise_core_start_index":
            0,

        "exercise_core_end_index":
            1,

        "first_nonexercise_index":
            None,

        "basis":
            (
                "The sole narrative record is the complete exercise "
                "requiring interviews with parents about allowance."
            ),
    },


    (
        "book:trainer-training-financial-literacy",
        "book:trainer-training-financial-literacy:fa-source",
        (
            "کار کودکان",
            "تمرین 10",
        ),
    ): {
        "decision":
            "FULL_GROUP_IS_EXERCISE_UNIT",

        "exercise_core_start_index":
            0,

        "exercise_core_end_index":
            1,

        "first_nonexercise_index":
            None,

        "basis":
            (
                "The sole narrative record is the complete exercise "
                "for designing age-appropriate income-generating activities."
            ),
    },


    (
        "book:trainer-training-financial-literacy",
        "book:trainer-training-financial-literacy:fa-source",
        (
            "کسب و کار کودکان",
            "تمرین 11",
        ),
    ): {
        "decision":
            "FULL_GROUP_IS_EXERCISE_UNIT",

        "exercise_core_start_index":
            0,

        "exercise_core_end_index":
            1,

        "first_nonexercise_index":
            None,

        "basis":
            (
                "The sole narrative record is the complete business-design "
                "exercise including production, costing, pricing and profit."
            ),
    },
}


# ============================================================
# 2) Resolve adjudications against actual Cell 40 rows
# ============================================================

additional_unresolved_adjudication_rows = []


for row in (
    additional_unresolved_boundary_rows
):

    key = (
        row[
            "work_id"
        ],
        row[
            "edition_id"
        ],
        tuple(
            row[
                "heading_path"
            ]
        ),
    )


    decision = (
        ADDITIONAL_UNRESOLVED_BOUNDARY_ADJUDICATION.get(
            key
        )
    )


    if (
        decision
        is None
    ):

        raise AssertionError(
            "Missing Cell 41 adjudication for unresolved unit: "
            f"{key}"
        )


    start_index = (
        decision[
            "exercise_core_start_index"
        ]
    )

    end_index = (
        decision[
            "exercise_core_end_index"
        ]
    )

    first_nonexercise_index = (
        decision[
            "first_nonexercise_index"
        ]
    )


    full_sequence = (
        row[
            "full_sequence"
        ]
    )


    assert (
        0
        <= start_index
        <= end_index
        < len(
            full_sequence
        )
    )


    exercise_core_items = (
        full_sequence[
            start_index:
            end_index + 1
        ]
    )


    additional_unresolved_adjudication_rows.append(
        {
            "work_id":
                row[
                    "work_id"
                ],

            "edition_id":
                row[
                    "edition_id"
                ],

            "heading_path":
                row[
                    "heading_path"
                ],

            "structure_group_key":
                row[
                    "structure_group_key"
                ],

            "decision":
                decision[
                    "decision"
                ],

            "basis":
                decision[
                    "basis"
                ],

            "exercise_core_start_index":
                start_index,

            "exercise_core_end_index":
                end_index,

            "first_nonexercise_index":
                first_nonexercise_index,

            "exercise_core_document_ids":
                [
                    item[
                        "document_id"
                    ]
                    for item
                    in exercise_core_items
                ],

            "exercise_core_record_count":
                len(
                    exercise_core_items
                ),

            "exercise_core_token_count":
                sum(
                    item[
                        "token_count"
                    ]
                    for item
                    in exercise_core_items
                ),

            "boundary_status":
                "ADJUDICATED_NOT_APPLIED",
        }
    )


# ============================================================
# 3) Validate the 17 adjudications independently
# ============================================================

additional_adjudication_integrity_failures = []


for adjudicated_row in (
    additional_unresolved_adjudication_rows
):

    source_row = next(
        row
        for row
        in additional_unresolved_boundary_rows
        if (
            row[
                "structure_group_key"
            ]
            ==
            adjudicated_row[
                "structure_group_key"
            ]
        )
    )


    full_sequence = (
        source_row[
            "full_sequence"
        ]
    )


    expected_ids = [
        item[
            "document_id"
        ]
        for item
        in full_sequence[
            adjudicated_row[
                "exercise_core_start_index"
            ]:
            adjudicated_row[
                "exercise_core_end_index"
            ] + 1
        ]
    ]


    if (
        expected_ids
        !=
        adjudicated_row[
            "exercise_core_document_ids"
        ]
    ):

        additional_adjudication_integrity_failures.append(
            {
                "heading_path":
                    adjudicated_row[
                        "heading_path"
                    ],

                "reason":
                    "CORE_DOCUMENT_SEQUENCE_MISMATCH",
            }
        )


    if (
        adjudicated_row[
            "decision"
        ]
        ==
        "FULL_GROUP_IS_EXERCISE_UNIT"
    ):

        if not (
            adjudicated_row[
                "exercise_core_start_index"
            ]
            == 0

            and

            adjudicated_row[
                "exercise_core_end_index"
            ]
            ==
            len(
                full_sequence
            )
            - 1

            and

            adjudicated_row[
                "first_nonexercise_index"
            ]
            is None
        ):

            additional_adjudication_integrity_failures.append(
                {
                    "heading_path":
                        adjudicated_row[
                            "heading_path"
                        ],

                    "reason":
                        "FULL_GROUP_DECISION_NOT_FULL_GROUP",
                }
            )


    if (
        adjudicated_row[
            "decision"
        ]
        ==
        "INTERNAL_CONTENT_TRANSITION_BOUNDARY"
    ):

        if not (
            adjudicated_row[
                "first_nonexercise_index"
            ]
            ==
            adjudicated_row[
                "exercise_core_end_index"
            ]
            + 1
        ):

            additional_adjudication_integrity_failures.append(
                {
                    "heading_path":
                        adjudicated_row[
                            "heading_path"
                        ],

                    "reason":
                        "INTERNAL_TRANSITION_NOT_CONTIGUOUS",
                }
            )


# ============================================================
# 4) Build resolved boundary rows for the 10 new explicit
# units
# ============================================================

additional_explicit_resolved_boundary_rows = []


for row in (
    additional_explicit_boundary_rows
):

    additional_explicit_resolved_boundary_rows.append(
        {
            "work_id":
                row[
                    "work_id"
                ],

            "edition_id":
                row[
                    "edition_id"
                ],

            "heading_path":
                row[
                    "heading_path"
                ],

            "structure_group_key":
                row[
                    "structure_group_key"
                ],

            "boundary_source":
                "EXPLICIT_SUPPLEMENTAL_MARKER",

            "exercise_core_start_index":
                row[
                    "exercise_core_start_index"
                ],

            "exercise_core_end_index":
                row[
                    "exercise_core_end_index"
                ],

            "first_nonexercise_index":
                row[
                    "supplemental_start_index"
                ],

            "exercise_core_document_ids":
                list(
                    row[
                        "candidate_core_document_ids"
                    ]
                ),

            "exercise_core_record_count":
                row[
                    "candidate_core_record_count"
                ],

            "exercise_core_token_count":
                row[
                    "candidate_core_token_count"
                ],

            "boundary_status":
                "BOUNDARY_RESOLVED_NOT_APPLIED",
        }
    )


# ============================================================
# 5) Build resolved rows for the 17 adjudicated units
# ============================================================

additional_adjudicated_resolved_boundary_rows = []


for row in (
    additional_unresolved_adjudication_rows
):

    additional_adjudicated_resolved_boundary_rows.append(
        {
            "work_id":
                row[
                    "work_id"
                ],

            "edition_id":
                row[
                    "edition_id"
                ],

            "heading_path":
                row[
                    "heading_path"
                ],

            "structure_group_key":
                row[
                    "structure_group_key"
                ],

            "boundary_source":
                row[
                    "decision"
                ],

            "exercise_core_start_index":
                row[
                    "exercise_core_start_index"
                ],

            "exercise_core_end_index":
                row[
                    "exercise_core_end_index"
                ],

            "first_nonexercise_index":
                row[
                    "first_nonexercise_index"
                ],

            "exercise_core_document_ids":
                list(
                    row[
                        "exercise_core_document_ids"
                    ]
                ),

            "exercise_core_record_count":
                row[
                    "exercise_core_record_count"
                ],

            "exercise_core_token_count":
                row[
                    "exercise_core_token_count"
                ],

            "boundary_status":
                "BOUNDARY_RESOLVED_NOT_APPLIED",
        }
    )


# ============================================================
# 6) Build complete 45-unit boundary contract
#
# 18 old resolved
# + 10 additional explicit
# + 17 additional adjudicated
# ============================================================

full_45_exercise_boundary_contract_rows = []


# ------------------------------------------------------------
# Existing 18
# ------------------------------------------------------------

for row in (
    exercise_boundary_contract_rows
):

    full_45_exercise_boundary_contract_rows.append(
        {
            "work_id":
                row[
                    "work_id"
                ],

            "edition_id":
                row[
                    "edition_id"
                ],

            "heading_path":
                list(
                    row[
                        "heading_path"
                    ]
                ),

            "structure_group_key":
                row[
                    "structure_group_key"
                ],

            "boundary_source":
                row[
                    "boundary_source"
                ],

            "exercise_core_start_index":
                row[
                    "exercise_core_start_index"
                ],

            "exercise_core_end_index":
                row[
                    "exercise_core_end_index"
                ],

            "first_nonexercise_index":
                row[
                    "first_nonexercise_index"
                ],

            "exercise_core_document_ids":
                list(
                    row[
                        "exercise_core_document_ids"
                    ]
                ),

            "exercise_core_record_count":
                row[
                    "exercise_core_record_count"
                ],

            "exercise_core_token_count":
                row[
                    "exercise_core_token_count"
                ],

            "discovery_origin":
                "LEGACY_SELECTED_SAMPLE_RECONSTRUCTION",

            "boundary_status":
                "BOUNDARY_RESOLVED_NOT_APPLIED",
        }
    )


# ------------------------------------------------------------
# Additional 10 explicit
# ------------------------------------------------------------

for row in (
    additional_explicit_resolved_boundary_rows
):

    full_45_exercise_boundary_contract_rows.append(
        {
            **row,

            "discovery_origin":
                "FULL_CORPUS_ADDITIONAL_DISCOVERY",
        }
    )


# ------------------------------------------------------------
# Additional 17 adjudicated
# ------------------------------------------------------------

for row in (
    additional_adjudicated_resolved_boundary_rows
):

    full_45_exercise_boundary_contract_rows.append(
        {
            **row,

            "discovery_origin":
                "FULL_CORPUS_ADDITIONAL_DISCOVERY",
        }
    )


# ============================================================
# 7) Complete contract accounting
# ============================================================

full_45_boundary_source_counts = Counter(
    row[
        "boundary_source"
    ]
    for row
    in full_45_exercise_boundary_contract_rows
)


full_45_discovery_origin_counts = Counter(
    row[
        "discovery_origin"
    ]
    for row
    in full_45_exercise_boundary_contract_rows
)


full_45_structure_keys = {
    row[
        "structure_group_key"
    ]
    for row
    in full_45_exercise_boundary_contract_rows
}


full_45_core_record_counts = [
    row[
        "exercise_core_record_count"
    ]
    for row
    in full_45_exercise_boundary_contract_rows
]


full_45_core_token_counts = [
    row[
        "exercise_core_token_count"
    ]
    for row
    in full_45_exercise_boundary_contract_rows
]


# ============================================================
# 8) Validate exact coverage against Cell 39 broad universe
# ============================================================

broad_universe_keys = {
    row[
        "structure_group_key"
    ]
    for row
    in full_exercise_like_group_rows
}


boundary_contract_missing_broad_units = sorted(
    broad_universe_keys
    -
    full_45_structure_keys,
    key=str,
)


boundary_contract_extra_units = sorted(
    full_45_structure_keys
    -
    broad_universe_keys,
    key=str,
)


# ============================================================
# 9) Core structural / corpus integrity
# ============================================================

full_45_boundary_integrity_failures = []

cross_unit_core_document_usage = defaultdict(
    list
)


for row in (
    full_45_exercise_boundary_contract_rows
):

    group_records = (
        book_heading_groups[
            row[
                "structure_group_key"
            ]
        ]
    )


    expected_core_ids = [
        record[
            "document_id"
        ]
        for record
        in group_records[
            row[
                "exercise_core_start_index"
            ]:
            row[
                "exercise_core_end_index"
            ] + 1
        ]
    ]


    if (
        expected_core_ids
        !=
        row[
            "exercise_core_document_ids"
        ]
    ):

        full_45_boundary_integrity_failures.append(
            {
                "heading_path":
                    row[
                        "heading_path"
                    ],

                "reason":
                    "CORE_SEQUENCE_MISMATCH",
            }
        )


    if not (
        row[
            "exercise_core_document_ids"
        ]
    ):

        full_45_boundary_integrity_failures.append(
            {
                "heading_path":
                    row[
                        "heading_path"
                    ],

                "reason":
                    "EMPTY_EXERCISE_CORE",
            }
        )


    for document_id in (
        row[
            "exercise_core_document_ids"
        ]
    ):

        cross_unit_core_document_usage[
            document_id
        ].append(
            row[
                "structure_group_key"
            ]
        )


        if (
            document_id
            not in
            corpus_target_index
        ):

            full_45_boundary_integrity_failures.append(
                {
                    "heading_path":
                        row[
                            "heading_path"
                    ],

                    "document_id":
                        document_id,

                    "reason":
                        "DOCUMENT_NOT_IN_CORPUS",
                }
            )

            continue


        record = (
            corpus_target_index[
                document_id
            ]
        )


        if (
            record[
                "source_type"
            ]
            != "book"
        ):

            full_45_boundary_integrity_failures.append(
                {
                    "heading_path":
                        row[
                            "heading_path"
                        ],

                    "document_id":
                        document_id,

                    "reason":
                        "NON_BOOK_EVIDENCE",
                }
            )


        if not (
            record[
                "text"
            ]
            or ""
        ).strip():

            full_45_boundary_integrity_failures.append(
                {
                    "heading_path":
                        row[
                            "heading_path"
                        ],

                    "document_id":
                        document_id,

                    "reason":
                        "EMPTY_TEXT",
                }
            )


        if not (
            record[
                "locator"
            ]
        ):

            full_45_boundary_integrity_failures.append(
                {
                    "heading_path":
                        row[
                            "heading_path"
                        ],

                    "document_id":
                        document_id,

                    "reason":
                        "MISSING_LOCATOR",
                }
            )


        if (
            record[
                "lineage"
            ]
            is None
        ):

            full_45_boundary_integrity_failures.append(
                {
                    "heading_path":
                        row[
                            "heading_path"
                        ],

                    "document_id":
                        document_id,

                    "reason":
                        "MISSING_LINEAGE",
                }
            )


# ============================================================
# 10) Cross-unit core overlap
# ============================================================

full_45_cross_unit_core_overlap = {
    document_id:
        group_keys

    for document_id, group_keys
    in cross_unit_core_document_usage.items()

    if (
        len(
            set(
                group_keys
            )
        )
        > 1
    )
}


# ============================================================
# 11) Boundary Contract Registry
# ============================================================

FULL_45_EXERCISE_BOUNDARY_CONTRACT_V1_DRAFT = {
    "version":
        "full_45_exercise_boundary_contract_v1_draft",

    "status":
        "ALL_45_BROAD_EXERCISE_BOUNDARIES_RESOLVED_NOT_APPLIED",

    "broad_unit_count":
        len(
            full_45_exercise_boundary_contract_rows
        ),

    "legacy_sample_origin_count":
        full_45_discovery_origin_counts[
            "LEGACY_SELECTED_SAMPLE_RECONSTRUCTION"
        ],

    "additional_full_corpus_origin_count":
        full_45_discovery_origin_counts[
            "FULL_CORPUS_ADDITIONAL_DISCOVERY"
        ],

    "boundary_source_counts":
        dict(
            full_45_boundary_source_counts
        ),

    "eligibility_status":
        "NOT_YET_EVALUATED",

    "sampling_rebuild_status":
        "NOT_STARTED",

    "case_supersession_status":
        "NOT_STARTED",

    "target_repair_status":
        "NOT_STARTED",

    "question_generation_status":
        "NOT_STARTED",

    "expected_answer_generation_status":
        "NOT_STARTED",

    "artifact_write_status":
        "NONE",
}


# ============================================================
# 12) Validation gates
# ============================================================

exact_17_adjudicated_gate = (
    len(
        additional_unresolved_adjudication_rows
    )
    == 17
)


sixteen_full_group_decisions_gate = (
    sum(
        1
        for row
        in additional_unresolved_adjudication_rows
        if (
            row[
                "decision"
            ]
            ==
            "FULL_GROUP_IS_EXERCISE_UNIT"
        )
    )
    == 16
)


one_internal_transition_gate = (
    sum(
        1
        for row
        in additional_unresolved_adjudication_rows
        if (
            row[
                "decision"
            ]
            ==
            "INTERNAL_CONTENT_TRANSITION_BOUNDARY"
        )
    )
    == 1
)


additional_adjudication_integrity_gate = (
    len(
        additional_adjudication_integrity_failures
    )
    == 0
)


exact_45_boundary_contract_gate = (
    len(
        full_45_exercise_boundary_contract_rows
    )
    == 45
)


all_45_unique_gate = (
    len(
        full_45_structure_keys
    )
    == 45
)


exact_broad_universe_coverage_gate = (
    len(
        boundary_contract_missing_broad_units
    )
    == 0

    and

    len(
        boundary_contract_extra_units
    )
    == 0

    and

    full_45_structure_keys
    ==
    broad_universe_keys
)


all_45_boundary_integrity_gate = (
    len(
        full_45_boundary_integrity_failures
    )
    == 0
)


no_cross_unit_core_overlap_gate = (
    len(
        full_45_cross_unit_core_overlap
    )
    == 0
)


eligibility_not_started_gate = (
    FULL_45_EXERCISE_BOUNDARY_CONTRACT_V1_DRAFT[
        "eligibility_status"
    ]
    ==
    "NOT_YET_EVALUATED"
)


sampling_not_started_gate = (
    FULL_45_EXERCISE_BOUNDARY_CONTRACT_V1_DRAFT[
        "sampling_rebuild_status"
    ]
    ==
    "NOT_STARTED"
)


supersession_not_started_gate = (
    FULL_45_EXERCISE_BOUNDARY_CONTRACT_V1_DRAFT[
        "case_supersession_status"
    ]
    ==
    "NOT_STARTED"
)


target_repair_not_started_gate = (
    FULL_45_EXERCISE_BOUNDARY_CONTRACT_V1_DRAFT[
        "target_repair_status"
    ]
    ==
    "NOT_STARTED"
)


question_generation_not_started_gate = (
    FULL_45_EXERCISE_BOUNDARY_CONTRACT_V1_DRAFT[
        "question_generation_status"
    ]
    ==
    "NOT_STARTED"
)


expected_answer_not_started_gate = (
    FULL_45_EXERCISE_BOUNDARY_CONTRACT_V1_DRAFT[
        "expected_answer_generation_status"
    ]
    ==
    "NOT_STARTED"
)


no_artifact_write_gate = (
    FULL_45_EXERCISE_BOUNDARY_CONTRACT_V1_DRAFT[
        "artifact_write_status"
    ]
    ==
    "NONE"
)


cell_41_gates = {
    "exact_17_additional_unresolved_units_adjudicated":
        exact_17_adjudicated_gate,

    "exact_16_full_group_exercise_decisions":
        sixteen_full_group_decisions_gate,

    "exact_1_internal_content_transition_boundary":
        one_internal_transition_gate,

    "all_17_adjudications_structurally_valid":
        additional_adjudication_integrity_gate,

    "exact_45_exercise_like_boundaries_resolved":
        exact_45_boundary_contract_gate,

    "all_45_structural_units_unique":
        all_45_unique_gate,

    "45_boundary_contract_exactly_covers_cell39_broad_universe":
        exact_broad_universe_coverage_gate,

    "all_45_exercise_cores_preserve_source_structure_and_lineage":
        all_45_boundary_integrity_gate,

    "no_document_shared_between_exercise_cores":
        no_cross_unit_core_overlap_gate,

    "final_unit_eligibility_not_yet_evaluated":
        eligibility_not_started_gate,

    "sampling_rebuild_not_started":
        sampling_not_started_gate,

    "case_supersession_not_started":
        supersession_not_started_gate,

    "target_repair_not_started":
        target_repair_not_started_gate,

    "question_generation_not_started":
        question_generation_not_started_gate,

    "expected_answer_generation_not_started":
        expected_answer_not_started_gate,

    "no_artifact_write":
        no_artifact_write_gate,
}


cell_41_pass = all(
    cell_41_gates.values()
)


# ============================================================
# 13) Output
# ============================================================

print(
    "=== ADDITIONAL 17 EXERCISE BOUNDARY ADJUDICATION ==="
)

print(
    "Adjudicated units:",
    len(
        additional_unresolved_adjudication_rows
    )
)

print(
    "Full-group units:",
    sum(
        1
        for row
        in additional_unresolved_adjudication_rows
        if (
            row[
                "decision"
            ]
            ==
            "FULL_GROUP_IS_EXERCISE_UNIT"
        )
    )
)

print(
    "Internal-transition units:",
    sum(
        1
        for row
        in additional_unresolved_adjudication_rows
        if (
            row[
                "decision"
            ]
            ==
            "INTERNAL_CONTENT_TRANSITION_BOUNDARY"
        )
    )
)


print(
    "\n=== ADJUDICATED 17 UNITS ==="
)

for row in (
    additional_unresolved_adjudication_rows
):

    print(
        "\nHeading:",
        row[
            "heading_path"
        ]
    )

    print(
        "Decision:",
        row[
            "decision"
        ]
    )

    print(
        "Core indices:",
        (
            row[
                "exercise_core_start_index"
            ],
            row[
                "exercise_core_end_index"
            ],
        )
    )

    print(
        "First non-exercise index:",
        row[
            "first_nonexercise_index"
        ]
    )

    print(
        "Core records:",
        row[
            "exercise_core_record_count"
        ],
        "| tokens:",
        row[
            "exercise_core_token_count"
        ],
    )

    print(
        "Basis:",
        row[
            "basis"
        ]
    )


print(
    "\n=== COMPLETE 45-UNIT BOUNDARY CONTRACT ==="
)

print(
    "Total broad exercise-like units:",
    len(
        full_45_exercise_boundary_contract_rows
    )
)

print(
    "Discovery origins:",
    dict(
        full_45_discovery_origin_counts
    )
)

print(
    "Boundary sources:",
    dict(
        full_45_boundary_source_counts
    )
)


print(
    "\nExercise-core record-count distribution:",
    {
        "min":
            min(
                full_45_core_record_counts
            ),

        "median":
            empirical_nearest_rank(
                full_45_core_record_counts,
                0.50,
            ),

        "max":
            max(
                full_45_core_record_counts
            ),
    }
)


print(
    "Exercise-core token-count distribution:",
    {
        "min":
            min(
                full_45_core_token_counts
            ),

        "median":
            empirical_nearest_rank(
                full_45_core_token_counts,
                0.50,
            ),

        "max":
            max(
                full_45_core_token_counts
            ),
    }
)


print(
    "\nMissing broad-universe units:",
    len(
        boundary_contract_missing_broad_units
    )
)

print(
    "Extra boundary-contract units:",
    len(
        boundary_contract_extra_units
    )
)

print(
    "Boundary integrity failures:",
    len(
        full_45_boundary_integrity_failures
    )
)

print(
    "Cross-unit core document overlap:",
    len(
        full_45_cross_unit_core_overlap
    )
)


print(
    "\n=== CODE CELL 41 GATES ==="
)

for name, passed in (
    cell_41_gates.items()
):

    print(
        name,
        ":",
        "PASS"
        if passed
        else "FAIL"
    )


print(
    "\nCODE CELL 41 GATE:",
    "PASS"
    if cell_41_pass
    else "FAIL"
)


assert cell_41_pass, (
    "Full 45-unit Exercise Boundary Contract failed. "
    "Do not evaluate eligibility or rebuild sampling until "
    "all 45 broad exercise-like structural units have complete, "
    "non-overlapping and source-valid boundaries."
)
```

<div style="direction: rtl; white-space: normal; line-height: 1.7;">

<h3>Exercise Unit Eligibility Audit</h3>

<p>
Cell 41 یک Boundary Contract کامل برای ۴۵ Exercise-like Unit ساخت.
اما «Exercise-like» هنوز به معنای عضویت نهایی در Population این Stratum نیست.
</p>

<p>
در این Cell هر Unit بر اساس معیارهای صریح و بازتولیدپذیر ارزیابی می‌شود.
Eligibility در این مرحله به معنی
<strong>واجد شرایط بودن برای Population سطح Exercise</strong>
است، نه Final Golden Membership.
</p>

<h4>معیارهای Eligibility</h4>

<ul>
<li>
آخرین جزء Heading Path باید دارای Marker صریح
<code>کاربرگ</code>،
<code>خودارزیابی</code>،
<code>خودآزمایی</code>،
<code>تمرین</code>
یا Marker پذیرفته‌شده‌ی Exercise باشد.
</li>

<li>
Exercise Core باید Boundary حل‌شده و غیرخالی داشته باشد.
</li>

<li>
Core باید علاوه بر Heading حداقل یک Record غیر-heading با متن واقعی داشته باشد.
</li>

<li>
تمام Evidence Documentها باید در Accepted Corpus وجود داشته باشند.
</li>

<li>
Locator، Lineage، Attribution و Rights باید حفظ شده باشند.
</li>

<li>
Core نباید با Exercise Unit دیگری Document مشترک داشته باشد.
</li>

<li>
هیچ Supplemental Record بعد از Boundary نباید وارد Core شده باشد.
</li>
</ul>

<p>
هیچ Unit به دلیل طول کم یا زیاد Token رد نمی‌شود.
Token Count در این مرحله Eligibility Rule نیست.
</p>

<p>
همچنین هیچ Case Sampling یا Target ساخته نمی‌شود.
خروجی فقط Population Eligibility Audit است.
</p>

</div>

```python
# Code Cell 42

# ============================================================
# Exercise Unit Eligibility Audit
#
# Input:
#   full_45_exercise_boundary_contract_rows
#
# Goal:
#   determine which resolved exercise units are eligible for
#   the rebuilt exercise_or_self_assessment population.
#
# Eligibility != final Golden membership.
#
# NO sampling.
# NO supersession.
# NO Target Evidence creation.
# NO question generation.
# NO artifact writes.
# ============================================================


# ============================================================
# 1) Eligibility contract
# ============================================================

EXERCISE_UNIT_ELIGIBILITY_CONTRACT_V1_DRAFT = {
    "version":
        "exercise_unit_eligibility_contract_v1_draft",

    "stratum":
        "exercise_or_self_assessment",

    "identity_level":
        "EXERCISE_UNIT",

    "required_conditions": [
        "TERMINAL_EXERCISE_MARKER",
        "BOUNDARY_RESOLVED",
        "NONEMPTY_CORE",
        "AT_LEAST_ONE_NONHEADING_CORE_RECORD",
        "AT_LEAST_ONE_NONEMPTY_NONHEADING_TEXT",
        "ALL_CORE_DOCUMENTS_IN_ACCEPTED_CORPUS",
        "LOCATOR_PRESENT",
        "LINEAGE_PRESENT",
        "ATTRIBUTION_PRESENT",
        "RIGHTS_PRESENT",
        "NO_CROSS_UNIT_DOCUMENT_OVERLAP",
        "NO_POST_BOUNDARY_DOCUMENT_IN_CORE",
    ],

    "token_threshold":
        None,

    "minimum_token_rule":
        False,

    "maximum_token_rule":
        False,

    "sampling_status":
        "NOT_STARTED",

    "golden_membership_status":
        "NOT_STARTED",

    "legacy_supersession_status":
        "NOT_STARTED",

    "target_repair_status":
        "NOT_STARTED",

    "question_generation_status":
        "NOT_STARTED",

    "expected_answer_generation_status":
        "NOT_STARTED",

    "artifact_write_status":
        "NONE",
}


# ============================================================
# 2) Independent cross-unit document ownership map
# ============================================================

exercise_core_document_owner_map = defaultdict(
    list
)


for unit_row in (
    full_45_exercise_boundary_contract_rows
):

    for document_id in (
        unit_row[
            "exercise_core_document_ids"
        ]
    ):

        exercise_core_document_owner_map[
            document_id
        ].append(
            unit_row[
                "structure_group_key"
            ]
        )


exercise_core_shared_documents = {
    document_id:
        group_keys

    for document_id, group_keys
    in exercise_core_document_owner_map.items()

    if (
        len(
            set(
                group_keys
            )
        )
        > 1
    )
}


# ============================================================
# 3) Evaluate all 45 units
# ============================================================

exercise_unit_eligibility_rows = []


for unit_row in (
    full_45_exercise_boundary_contract_rows
):

    group_key = (
        unit_row[
            "structure_group_key"
        ]
    )


    group_records = (
        book_heading_groups[
            group_key
        ]
    )


    heading_path = list(
        unit_row[
            "heading_path"
        ]
    )


    terminal_heading = (
        heading_path[
            -1
        ]
        if heading_path
        else ""
    )


    terminal_marker_hits = (
        exercise_heading_marker_hits(
            [
                terminal_heading
            ]
        )
    )


    core_document_ids = list(
        unit_row[
            "exercise_core_document_ids"
        ]
    )


    core_document_id_set = set(
        core_document_ids
    )


    core_records = [
        corpus_target_index[
            document_id
        ]
        for document_id
        in core_document_ids
    ]


    # --------------------------------------------------------
    # Structural checks
    # --------------------------------------------------------

    terminal_marker_gate = (
        len(
            terminal_marker_hits
        )
        > 0
    )


    boundary_resolved_gate = (
        unit_row[
            "boundary_status"
        ]
        ==
        "BOUNDARY_RESOLVED_NOT_APPLIED"
    )


    nonempty_core_gate = (
        len(
            core_records
        )
        > 0
    )


    # --------------------------------------------------------
    # Non-heading evidence
    # --------------------------------------------------------

    nonheading_core_records = []


    for document_id in (
        core_document_ids
    ):

        source_group_record = next(
            record
            for record
            in group_records
            if (
                record[
                    "document_id"
                ]
                ==
                document_id
            )
        )


        if (
            source_group_record[
                "is_heading"
            ]
            is not True
        ):

            nonheading_core_records.append(
                source_group_record
            )


    nonheading_record_gate = (
        len(
            nonheading_core_records
        )
        >= 1
    )


    nonempty_nonheading_text_gate = any(
        bool(
            (
                record[
                    "text"
                ]
                or ""
            ).strip()
        )

        for record
        in nonheading_core_records
    )


    # --------------------------------------------------------
    # Corpus / metadata integrity
    # --------------------------------------------------------

    all_documents_in_corpus_gate = all(
        document_id
        in
        corpus_target_index

        for document_id
        in core_document_ids
    )


    locator_gate = all(
        bool(
            corpus_target_index[
                document_id
            ][
                "locator"
            ]
        )

        for document_id
        in core_document_ids
    )


    lineage_gate = all(
        corpus_target_index[
            document_id
        ][
            "lineage"
        ]
        is not None

        for document_id
        in core_document_ids
    )


    attribution_gate = all(
        corpus_target_index[
            document_id
        ][
            "attribution"
        ]
        is not None

        for document_id
        in core_document_ids
    )


    rights_gate = all(
        corpus_target_index[
            document_id
        ][
            "rights"
        ]
        is not None

        for document_id
        in core_document_ids
    )


    # --------------------------------------------------------
    # Cross-unit overlap
    # --------------------------------------------------------

    no_cross_unit_overlap_gate = all(
        len(
            set(
                exercise_core_document_owner_map[
                    document_id
                ]
            )
        )
        == 1

        for document_id
        in core_document_ids
    )


    # --------------------------------------------------------
    # Validate exact Boundary:
    #
    # no record after first_nonexercise_index may be present
    # inside core_document_ids.
    # --------------------------------------------------------

    first_nonexercise_index = (
        unit_row[
            "first_nonexercise_index"
        ]
    )


    if (
        first_nonexercise_index
        is None
    ):

        no_post_boundary_document_gate = (
            unit_row[
                "exercise_core_end_index"
            ]
            ==
            len(
                group_records
            )
            - 1
        )


    else:

        post_boundary_document_ids = {
            record[
                "document_id"
            ]
            for record
            in group_records[
                first_nonexercise_index:
            ]
        }


        no_post_boundary_document_gate = (
            len(
                core_document_id_set
                &
                post_boundary_document_ids
            )
            == 0

            and

            unit_row[
                "exercise_core_end_index"
            ]
            ==
            first_nonexercise_index
            - 1
        )


    # --------------------------------------------------------
    # Unit-level eligibility conditions
    # --------------------------------------------------------

    eligibility_conditions = {
        "terminal_exercise_marker":
            terminal_marker_gate,

        "boundary_resolved":
            boundary_resolved_gate,

        "nonempty_core":
            nonempty_core_gate,

        "has_nonheading_core_record":
            nonheading_record_gate,

        "has_nonempty_nonheading_text":
            nonempty_nonheading_text_gate,

        "all_documents_in_accepted_corpus":
            all_documents_in_corpus_gate,

        "locator_present":
            locator_gate,

        "lineage_present":
            lineage_gate,

        "attribution_present":
            attribution_gate,

        "rights_present":
            rights_gate,

        "no_cross_unit_document_overlap":
            no_cross_unit_overlap_gate,

        "no_post_boundary_document_in_core":
            no_post_boundary_document_gate,
    }


    eligibility_pass = all(
        eligibility_conditions.values()
    )


    if eligibility_pass:

        eligibility_status = (
            "ELIGIBLE_FOR_EXERCISE_UNIT_POPULATION"
        )

    else:

        eligibility_status = (
            "INELIGIBLE_PENDING_REVIEW"
        )


    failed_conditions = [
        condition
        for condition, passed
        in eligibility_conditions.items()
        if (
            passed
            is False
        )
    ]


    exercise_unit_eligibility_rows.append(
        {
            "work_id":
                unit_row[
                    "work_id"
                ],

            "edition_id":
                unit_row[
                    "edition_id"
                ],

            "heading_path":
                heading_path,

            "structure_group_key":
                group_key,

            "boundary_source":
                unit_row[
                    "boundary_source"
                ],

            "terminal_heading":
                terminal_heading,

            "terminal_marker_hits":
                terminal_marker_hits,

            "exercise_core_document_ids":
                core_document_ids,

            "exercise_core_document_count":
                len(
                    core_document_ids
                ),

            "exercise_core_token_count":
                unit_row[
                    "exercise_core_token_count"
                ],

            "nonheading_core_record_count":
                len(
                    nonheading_core_records
                ),

            "eligibility_conditions":
                eligibility_conditions,

            "failed_conditions":
                failed_conditions,

            "eligibility_status":
                eligibility_status,

            "final_golden_membership":
                False,

            "sampling_status":
                "NOT_STARTED",
        }
    )


# ============================================================
# 4) Aggregate eligibility
# ============================================================

exercise_unit_eligibility_status_counts = Counter(
    row[
        "eligibility_status"
    ]
    for row
    in exercise_unit_eligibility_rows
)


eligible_exercise_unit_rows = [
    row
    for row
    in exercise_unit_eligibility_rows
    if (
        row[
            "eligibility_status"
        ]
        ==
        "ELIGIBLE_FOR_EXERCISE_UNIT_POPULATION"
    )
]


ineligible_exercise_unit_rows = [
    row
    for row
    in exercise_unit_eligibility_rows
    if (
        row[
            "eligibility_status"
        ]
        ==
        "INELIGIBLE_PENDING_REVIEW"
    )
]


# ============================================================
# 5) Failure-condition distribution
# ============================================================

exercise_unit_failed_condition_counts = Counter()


for row in (
    ineligible_exercise_unit_rows
):

    for condition in (
        row[
            "failed_conditions"
        ]
    ):

        exercise_unit_failed_condition_counts[
            condition
        ] += 1


# ============================================================
# 6) Eligible population structural distributions
# ============================================================

eligible_unit_record_counts = [
    row[
        "exercise_core_document_count"
    ]
    for row
    in eligible_exercise_unit_rows
]


eligible_unit_token_counts = [
    row[
        "exercise_core_token_count"
    ]
    for row
    in eligible_exercise_unit_rows
]


eligible_units_by_work = Counter(
    row[
        "work_id"
    ]
    for row
    in eligible_exercise_unit_rows
)


eligible_units_by_terminal_marker = Counter()


for row in (
    eligible_exercise_unit_rows
):

    unique_markers = {
        hit[
            "marker"
        ]
        for hit
        in row[
            "terminal_marker_hits"
        ]
    }


    for marker in (
        unique_markers
    ):

        eligible_units_by_terminal_marker[
            marker
        ] += 1


# ============================================================
# 7) Ensure eligibility is independent of token threshold
# ============================================================

token_rule_violation_rows = []


for row in (
    exercise_unit_eligibility_rows
):

    if (
        row[
            "exercise_core_token_count"
        ]
        <= 0
    ):

        token_rule_violation_rows.append(
            {
                "heading_path":
                    row[
                        "heading_path"
                    ],

                "reason":
                    "NONPOSITIVE_TOKEN_COUNT",
            }
        )


# ============================================================
# 8) Eligibility Registry
# ============================================================

EXERCISE_UNIT_ELIGIBILITY_AUDIT_V1_DRAFT = {
    "version":
        "exercise_unit_eligibility_audit_v1_draft",

    "status":
        "ELIGIBILITY_EVALUATED_NO_SAMPLING",

    "audited_unit_count":
        len(
            exercise_unit_eligibility_rows
        ),

    "eligible_population_count":
        len(
            eligible_exercise_unit_rows
        ),

    "ineligible_pending_review_count":
        len(
            ineligible_exercise_unit_rows
        ),

    "token_threshold_used":
        False,

    "population_freeze_status":
        "NOT_STARTED",

    "sampling_rebuild_status":
        "NOT_STARTED",

    "legacy_case_supersession_status":
        "NOT_STARTED",

    "target_repair_status":
        "NOT_STARTED",

    "question_generation_status":
        "NOT_STARTED",

    "expected_answer_generation_status":
        "NOT_STARTED",

    "artifact_write_status":
        "NONE",
}


# ============================================================
# 9) Validation gates
# ============================================================

exact_45_units_audited_gate = (
    len(
        exercise_unit_eligibility_rows
    )
    == 45
)


all_45_accounted_gate = (
    len(
        eligible_exercise_unit_rows
    )
    +
    len(
        ineligible_exercise_unit_rows
    )
    == 45
)


eligibility_conditions_complete_gate = all(
    len(
        row[
            "eligibility_conditions"
        ]
    )
    == 12

    for row
    in exercise_unit_eligibility_rows
)


eligibility_status_consistency_gate = all(
    (
        row[
            "eligibility_status"
        ]
        ==
        "ELIGIBLE_FOR_EXERCISE_UNIT_POPULATION"

        and

        len(
            row[
                "failed_conditions"
            ]
        )
        == 0
    )

    or

    (
        row[
            "eligibility_status"
        ]
        ==
        "INELIGIBLE_PENDING_REVIEW"

        and

        len(
            row[
                "failed_conditions"
            ]
        )
        > 0
    )

    for row
    in exercise_unit_eligibility_rows
)


no_cross_unit_overlap_gate = (
    len(
        exercise_core_shared_documents
    )
    == 0
)


no_token_threshold_gate = (
    EXERCISE_UNIT_ELIGIBILITY_AUDIT_V1_DRAFT[
        "token_threshold_used"
    ]
    is False

    and

    EXERCISE_UNIT_ELIGIBILITY_CONTRACT_V1_DRAFT[
        "token_threshold"
    ]
    is None
)


all_core_tokens_positive_gate = (
    len(
        token_rule_violation_rows
    )
    == 0
)


population_not_frozen_gate = (
    EXERCISE_UNIT_ELIGIBILITY_AUDIT_V1_DRAFT[
        "population_freeze_status"
    ]
    ==
    "NOT_STARTED"
)


sampling_not_started_gate = (
    EXERCISE_UNIT_ELIGIBILITY_AUDIT_V1_DRAFT[
        "sampling_rebuild_status"
    ]
    ==
    "NOT_STARTED"
)


supersession_not_started_gate = (
    EXERCISE_UNIT_ELIGIBILITY_AUDIT_V1_DRAFT[
        "legacy_case_supersession_status"
    ]
    ==
    "NOT_STARTED"
)


target_repair_not_started_gate = (
    EXERCISE_UNIT_ELIGIBILITY_AUDIT_V1_DRAFT[
        "target_repair_status"
    ]
    ==
    "NOT_STARTED"
)


question_generation_not_started_gate = (
    EXERCISE_UNIT_ELIGIBILITY_AUDIT_V1_DRAFT[
        "question_generation_status"
    ]
    ==
    "NOT_STARTED"
)


expected_answer_not_started_gate = (
    EXERCISE_UNIT_ELIGIBILITY_AUDIT_V1_DRAFT[
        "expected_answer_generation_status"
    ]
    ==
    "NOT_STARTED"
)


no_artifact_write_gate = (
    EXERCISE_UNIT_ELIGIBILITY_AUDIT_V1_DRAFT[
        "artifact_write_status"
    ]
    ==
    "NONE"
)


cell_42_gates = {
    "exact_45_exercise_units_audited":
        exact_45_units_audited_gate,

    "all_45_units_receive_eligibility_status":
        all_45_accounted_gate,

    "all_unit_eligibility_conditions_complete":
        eligibility_conditions_complete_gate,

    "eligibility_status_matches_condition_results":
        eligibility_status_consistency_gate,

    "no_cross_unit_exercise_core_document_overlap":
        no_cross_unit_overlap_gate,

    "no_token_threshold_used_for_eligibility":
        no_token_threshold_gate,

    "all_exercise_cores_have_positive_observed_tokens":
        all_core_tokens_positive_gate,

    "population_freeze_not_started":
        population_not_frozen_gate,

    "sampling_rebuild_not_started":
        sampling_not_started_gate,

    "legacy_case_supersession_not_started":
        supersession_not_started_gate,

    "target_repair_not_started":
        target_repair_not_started_gate,

    "question_generation_not_started":
        question_generation_not_started_gate,

    "expected_answer_generation_not_started":
        expected_answer_not_started_gate,

    "no_artifact_write":
        no_artifact_write_gate,
}


cell_42_pass = all(
    cell_42_gates.values()
)


# ============================================================
# 10) Output
# ============================================================

print(
    "=== EXERCISE UNIT ELIGIBILITY AUDIT ==="
)

print(
    "Units audited:",
    len(
        exercise_unit_eligibility_rows
    )
)

print(
    "Eligible for Exercise Unit population:",
    len(
        eligible_exercise_unit_rows
    )
)

print(
    "Ineligible / pending review:",
    len(
        ineligible_exercise_unit_rows
    )
)


print(
    "\n=== ELIGIBILITY STATUS COUNTS ==="
)

for status, count in sorted(
    exercise_unit_eligibility_status_counts.items()
):

    print(
        "-",
        status,
        ":",
        count,
    )


print(
    "\n=== FAILED CONDITION COUNTS ==="
)

if (
    exercise_unit_failed_condition_counts
):

    for condition, count in sorted(
        exercise_unit_failed_condition_counts.items()
    ):

        print(
            "-",
            condition,
            ":",
            count,
        )

else:

    print(
        "NONE"
    )


print(
    "\n=== ELIGIBLE POPULATION BY WORK ==="
)

for work_id, count in sorted(
    eligible_units_by_work.items()
):

    print(
        "-",
        work_id,
        ":",
        count,
    )


print(
    "\n=== ELIGIBLE POPULATION BY TERMINAL MARKER ==="
)

for marker, count in sorted(
    eligible_units_by_terminal_marker.items()
):

    print(
        "-",
        marker,
        ":",
        count,
    )


if (
    eligible_exercise_unit_rows
):

    print(
        "\nEligible unit core-record distribution:",
        {
            "min":
                min(
                    eligible_unit_record_counts
                ),

            "median":
                empirical_nearest_rank(
                    eligible_unit_record_counts,
                    0.50,
                ),

            "max":
                max(
                    eligible_unit_record_counts
                ),
        }
    )


    print(
        "Eligible unit core-token distribution:",
        {
            "min":
                min(
                    eligible_unit_token_counts
                ),

            "median":
                empirical_nearest_rank(
                    eligible_unit_token_counts,
                    0.50,
                ),

            "max":
                max(
                    eligible_unit_token_counts
                ),
        }
    )


print(
    "\n=== INELIGIBLE / REVIEW UNITS ==="
)

if (
    ineligible_exercise_unit_rows
):

    for row in (
        ineligible_exercise_unit_rows
    ):

        print(
            "-",
            row[
                "heading_path"
            ],
            "| failures:",
            row[
                "failed_conditions"
            ],
        )

else:

    print(
        "NONE"
    )


print(
    "\n=== TOKEN RULE AUDIT ==="
)

print(
    "Token threshold used:",
    EXERCISE_UNIT_ELIGIBILITY_AUDIT_V1_DRAFT[
        "token_threshold_used"
    ]
)

print(
    "Nonpositive core-token rows:",
    len(
        token_rule_violation_rows
    )
)


print(
    "\n=== CODE CELL 42 GATES ==="
)

for name, passed in (
    cell_42_gates.items()
):

    print(
        name,
        ":",
        "PASS"
        if passed
        else "FAIL"
    )


print(
    "\nCODE CELL 42 GATE:",
    "PASS"
    if cell_42_pass
    else "FAIL"
)


assert cell_42_pass, (
    "Exercise Unit eligibility audit failed. "
    "Do not freeze the rebuilt exercise population or rerun "
    "sampling until all 45 units have complete and internally "
    "consistent eligibility results."
)
```

<div style="direction: rtl; white-space: normal; line-height: 1.7;">

<h3>Freeze the Rebuilt Exercise-Unit Population</h3>

<p>
Cell 42 نشان داد هر ۴۵ Exercise Unit کشف‌شده واجد شرایط Population سطح
<code>exercise_or_self_assessment</code>
هستند.
</p>

<p>
اکنون می‌توان Population این Stratum را در سطح
<strong>Exercise Unit</strong>
به‌صورت یک Snapshot پایدار در حافظه Freeze کرد.
</p>

<p>
این Freeze هنوز Golden Sampling یا Final Golden Membership نیست.
هدف آن ایجاد یک Population ثابت و قابل بازتولید است که Sampling بعدی
دقیقاً روی همین ۴۵ Unit اجرا شود.
</p>

<h4>Population Identity</h4>

<p>
هویت هر Unit از این عناصر ساخته می‌شود:
</p>

<ul>
<li>Work ID</li>
<li>Edition ID</li>
<li>Heading Path</li>
<li>Boundary Contract Version</li>
<li>ترتیب دقیق Evidence Documentهای داخل Exercise Core</li>
</ul>

<p>
برای هر Unit یک Stable Population ID و یک Evidence Fingerprint مستقل ساخته می‌شود.
همچنین Mapping کامل ۳۸ Golden Case رکوردی قبلی به Population جدید حفظ می‌شود.
</p>

<p>
در این مرحله:
</p>

<ul>
<li>۴۵ Unit فقط Population Member هستند؛</li>
<li>هیچ Unit هنوز Final Golden Case نیست؛</li>
<li>هیچ Sampling اجرا نمی‌شود؛</li>
<li>۳۸ Case قدیمی Supersede نمی‌شوند؛</li>
<li>Target Evidence و Question Generation شروع نمی‌شود؛</li>
<li>هیچ Artifact روی دیسک نوشته نمی‌شود.</li>
</ul>

</div>

```python
# Code Cell 43

# ============================================================
# Exercise-Unit Population Freeze
#
# Input:
#   - 45 resolved boundary units
#   - 45/45 eligibility PASS
#
# Output:
#   - stable 45-unit in-memory population snapshot
#   - stable population IDs
#   - evidence fingerprints
#   - legacy 38-case traceability
#   - whole-population snapshot fingerprint
#
# NO sampling.
# NO final Golden membership.
# NO supersession.
# NO Target Evidence mutation.
# NO question generation.
# NO artifact writes.
# ============================================================


# ============================================================
# 1) Population schema
# ============================================================

EXERCISE_UNIT_POPULATION_SCHEMA_VERSION = (
    "exercise_unit_population_v1_draft"
)


EXERCISE_UNIT_POPULATION_FREEZE_CONTRACT_V1_DRAFT = {
    "version":
        "exercise_unit_population_freeze_contract_v1_draft",

    "stratum":
        "exercise_or_self_assessment",

    "source_type":
        "book",

    "identity_level":
        "EXERCISE_UNIT",

    "eligibility_source":
        EXERCISE_UNIT_ELIGIBILITY_AUDIT_V1_DRAFT[
            "version"
        ],

    "boundary_contract_source":
        FULL_45_EXERCISE_BOUNDARY_CONTRACT_V1_DRAFT[
            "version"
        ],

    "population_membership_rule":
        "ELIGIBLE_FOR_EXERCISE_UNIT_POPULATION",

    "sampling_status":
        "NOT_STARTED",

    "final_golden_membership_status":
        "NOT_STARTED",

    "legacy_case_supersession_status":
        "NOT_STARTED",

    "target_repair_status":
        "NOT_STARTED",

    "question_generation_status":
        "NOT_STARTED",

    "expected_answer_generation_status":
        "NOT_STARTED",

    "artifact_write_status":
        "NONE",
}


# ============================================================
# 2) Stable population-unit ID
# ============================================================

def build_exercise_population_unit_id(
    work_id,
    edition_id,
    heading_path,
    exercise_core_document_ids,
):

    identity_payload = {
        "schema_version":
            EXERCISE_UNIT_POPULATION_SCHEMA_VERSION,

        "source_type":
            "book",

        "stratum":
            "exercise_or_self_assessment",

        "identity_level":
            "EXERCISE_UNIT",

        "work_id":
            work_id,

        "edition_id":
            edition_id,

        "heading_path":
            list(
                heading_path
            ),

        "boundary_contract_version":
            FULL_45_EXERCISE_BOUNDARY_CONTRACT_V1_DRAFT[
                "version"
            ],

        "exercise_core_document_ids":
            list(
                exercise_core_document_ids
            ),
    }


    digest = hashlib.sha256(
        stable_json_key(
            identity_payload
        ).encode(
            "utf-8"
        )
    ).hexdigest()


    return (
        "exercise-population-unit:"
        +
        digest[
            :32
        ]
    )


# ============================================================
# 3) Eligibility lookup by exact structural group
# ============================================================

exercise_eligibility_by_group = {
    row[
        "structure_group_key"
    ]:
        row

    for row
    in exercise_unit_eligibility_rows
}


assert (
    len(
        exercise_eligibility_by_group
    )
    == 45
)


# ============================================================
# 4) Build frozen population records
# ============================================================

exercise_unit_population_records = []

exercise_unit_population_by_id = {}

exercise_unit_population_by_group = {}


for boundary_row in sorted(
    full_45_exercise_boundary_contract_rows,
    key=lambda row: (
        str(
            row[
                "work_id"
            ]
        ),
        str(
            row[
                "edition_id"
            ]
        ),
        tuple(
            row[
                "heading_path"
            ]
        ),
    ),
):

    group_key = (
        boundary_row[
            "structure_group_key"
        ]
    )


    eligibility_row = (
        exercise_eligibility_by_group[
            group_key
        ]
    )


    assert (
        eligibility_row[
            "eligibility_status"
        ]
        ==
        "ELIGIBLE_FOR_EXERCISE_UNIT_POPULATION"
    )


    core_document_ids = list(
        boundary_row[
            "exercise_core_document_ids"
        ]
    )


    # --------------------------------------------------------
    # Stable unit ID
    # --------------------------------------------------------

    population_unit_id = (
        build_exercise_population_unit_id(
            work_id=(
                boundary_row[
                    "work_id"
                ]
            ),
            edition_id=(
                boundary_row[
                    "edition_id"
                ]
            ),
            heading_path=(
                boundary_row[
                    "heading_path"
                ]
            ),
            exercise_core_document_ids=(
                core_document_ids
            ),
        )
    )


    # --------------------------------------------------------
    # Exact ordered evidence fingerprint
    # --------------------------------------------------------

    evidence_fingerprint_payload = []


    for document_id in (
        core_document_ids
    ):

        text = (
            corpus_target_index[
                document_id
            ][
                "text"
            ]
        )


        evidence_fingerprint_payload.append(
            {
                "document_id":
                    document_id,

                "text_sha256":
                    hashlib.sha256(
                        text.encode(
                            "utf-8"
                        )
                    ).hexdigest(),
            }
        )


    evidence_sha256 = hashlib.sha256(
        stable_json_key(
            evidence_fingerprint_payload
        ).encode(
            "utf-8"
        )
    ).hexdigest()


    # --------------------------------------------------------
    # Marker identity
    # --------------------------------------------------------

    terminal_markers = sorted(
        {
            hit[
                "marker"
            ]
            for hit
            in eligibility_row[
                "terminal_marker_hits"
            ]
        }
    )


    population_record = {
        "population_unit_id":
            population_unit_id,

        "schema_version":
            EXERCISE_UNIT_POPULATION_SCHEMA_VERSION,

        "source_type":
            "book",

        "stratum":
            "exercise_or_self_assessment",

        "identity_level":
            "EXERCISE_UNIT",

        "work_id":
            boundary_row[
                "work_id"
            ],

        "edition_id":
            boundary_row[
                "edition_id"
            ],

        "heading_path":
            list(
                boundary_row[
                    "heading_path"
                ]
            ),

        "terminal_markers":
            terminal_markers,

        "structure_group_key":
            group_key,

        "discovery_origin":
            boundary_row[
                "discovery_origin"
            ],

        "boundary_source":
            boundary_row[
                "boundary_source"
            ],

        "exercise_core_start_index":
            boundary_row[
                "exercise_core_start_index"
            ],

        "exercise_core_end_index":
            boundary_row[
                "exercise_core_end_index"
            ],

        "first_nonexercise_index":
            boundary_row[
                "first_nonexercise_index"
            ],

        "evidence_document_ids":
            core_document_ids,

        "evidence_document_count":
            len(
                core_document_ids
            ),

        "exercise_core_token_count":
            boundary_row[
                "exercise_core_token_count"
            ],

        "exercise_core_evidence_sha256":
            evidence_sha256,

        "eligibility_status":
            eligibility_row[
                "eligibility_status"
            ],

        "population_membership":
            True,

        "sampling_status":
            "NOT_STARTED",

        "final_golden_membership":
            False,

        "legacy_supersession_status":
            "NOT_APPLIED",

        "target_evidence_status":
            "NOT_STARTED",

        "question_generation_status":
            "NOT_STARTED",

        "expected_answer_generation_status":
            "NOT_STARTED",
    }


    if (
        population_unit_id
        in
        exercise_unit_population_by_id
    ):

        raise AssertionError(
            "Duplicate population unit ID: "
            f"{population_unit_id}"
        )


    if (
        group_key
        in
        exercise_unit_population_by_group
    ):

        raise AssertionError(
            "Duplicate structural group in population: "
            f"{group_key}"
        )


    exercise_unit_population_records.append(
        population_record
    )


    exercise_unit_population_by_id[
        population_unit_id
    ] = (
        population_record
    )


    exercise_unit_population_by_group[
        group_key
    ] = (
        population_record
    )


# ============================================================
# 5) Rebuild traceability:
#
# 18 reconstructed sample units
# -> 45-unit frozen population
# ============================================================

reconstructed_18_to_population_mapping = []


for proposed_case in (
    proposed_exercise_unit_cases
):

    group_key = (
        build_book_structure_group_key(
            work_id=(
                proposed_case[
                    "work_id"
                ]
            ),
            edition_id=(
                proposed_case[
                    "edition_id"
                ]
            ),
            heading_path=(
                proposed_case[
                    "heading_path"
                ]
            ),
        )
    )


    population_record = (
        exercise_unit_population_by_group[
            group_key
        ]
    )


    reconstructed_18_to_population_mapping.append(
        {
            "proposed_exercise_unit_case_id":
                proposed_case[
                    "proposed_exercise_unit_case_id"
                ],

            "population_unit_id":
                population_record[
                    "population_unit_id"
                ],

            "structure_group_key":
                group_key,

            "heading_path":
                population_record[
                    "heading_path"
                ],

            "mapping_status":
                "EXACT_STRUCTURAL_IDENTITY_MATCH",
        }
    )


# ============================================================
# 6) Legacy 38 Golden Case -> frozen 45-unit population
# ============================================================

proposed_case_id_to_population_id = {
    row[
        "proposed_exercise_unit_case_id"
    ]:
        row[
            "population_unit_id"
        ]

    for row
    in reconstructed_18_to_population_mapping
}


legacy_38_to_population_mapping = []


for legacy_row in (
    legacy_to_proposed_exercise_case_mapping
):

    proposed_case_id = (
        legacy_row[
            "proposed_exercise_unit_case_id"
        ]
    )


    population_unit_id = (
        proposed_case_id_to_population_id[
            proposed_case_id
        ]
    )


    legacy_38_to_population_mapping.append(
        {
            "legacy_golden_case_id":
                legacy_row[
                    "legacy_golden_case_id"
                ],

            "legacy_candidate_document_id":
                legacy_row[
                    "legacy_candidate_document_id"
                ],

            "legacy_candidate_position":
                legacy_row[
                    "legacy_candidate_position"
                ],

            "legacy_relationship":
                legacy_row[
                    "relationship"
                ],

            "population_unit_id":
                population_unit_id,

            "mapping_status":
                "TRACEABILITY_PRESERVED_NO_SUPERSESSION_APPLIED",
        }
    )


# ============================================================
# 7) Population-origin accounting
# ============================================================

exercise_population_origin_counts = Counter(
    record[
        "discovery_origin"
    ]
    for record
    in exercise_unit_population_records
)


exercise_population_work_counts = Counter(
    record[
        "work_id"
    ]
    for record
    in exercise_unit_population_records
)


exercise_population_marker_counts = Counter()


for record in (
    exercise_unit_population_records
):

    for marker in (
        record[
            "terminal_markers"
        ]
    ):

        exercise_population_marker_counts[
            marker
        ] += 1


legacy_represented_population_ids = {
    row[
        "population_unit_id"
    ]
    for row
    in legacy_38_to_population_mapping
}


population_units_without_legacy_case = [
    record
    for record
    in exercise_unit_population_records
    if (
        record[
            "population_unit_id"
        ]
        not in
        legacy_represented_population_ids
    )
]


# ============================================================
# 8) Independent stable-ID recomputation
# ============================================================

population_id_recomputation_failures = []


for record in (
    exercise_unit_population_records
):

    recomputed_id = (
        build_exercise_population_unit_id(
            work_id=(
                record[
                    "work_id"
                ]
            ),
            edition_id=(
                record[
                    "edition_id"
                ]
            ),
            heading_path=(
                record[
                    "heading_path"
                ]
            ),
            exercise_core_document_ids=(
                record[
                    "evidence_document_ids"
                ]
            ),
        )
    )


    if (
        recomputed_id
        !=
        record[
            "population_unit_id"
        ]
    ):

        population_id_recomputation_failures.append(
            record[
                "population_unit_id"
            ]
        )


# ============================================================
# 9) Independent evidence-fingerprint recomputation
# ============================================================

population_evidence_fingerprint_failures = []


for record in (
    exercise_unit_population_records
):

    fingerprint_payload = []


    for document_id in (
        record[
            "evidence_document_ids"
        ]
    ):

        text = (
            corpus_target_index[
                document_id
            ][
                "text"
            ]
        )


        fingerprint_payload.append(
            {
                "document_id":
                    document_id,

                "text_sha256":
                    hashlib.sha256(
                        text.encode(
                            "utf-8"
                        )
                    ).hexdigest(),
            }
        )


    recomputed_fingerprint = hashlib.sha256(
        stable_json_key(
            fingerprint_payload
        ).encode(
            "utf-8"
        )
    ).hexdigest()


    if (
        recomputed_fingerprint
        !=
        record[
            "exercise_core_evidence_sha256"
        ]
    ):

        population_evidence_fingerprint_failures.append(
            record[
                "population_unit_id"
            ]
        )


# ============================================================
# 10) Whole-population snapshot fingerprint
#
# Stable ordered snapshot.
# ============================================================

exercise_population_snapshot_payload = [
    {
        "population_unit_id":
            record[
                "population_unit_id"
            ],

        "work_id":
            record[
                "work_id"
            ],

        "edition_id":
            record[
                "edition_id"
            ],

        "heading_path":
            record[
                "heading_path"
            ],

        "boundary_source":
            record[
                "boundary_source"
            ],

        "evidence_document_ids":
            record[
                "evidence_document_ids"
            ],

        "exercise_core_evidence_sha256":
            record[
                "exercise_core_evidence_sha256"
            ],
    }

    for record
    in exercise_unit_population_records
]


exercise_population_snapshot_sha256 = hashlib.sha256(
    stable_json_key(
        exercise_population_snapshot_payload
    ).encode(
        "utf-8"
    )
).hexdigest()


# ============================================================
# 11) Cross-unit document overlap revalidation
# ============================================================

population_document_owner_map = defaultdict(
    list
)


for record in (
    exercise_unit_population_records
):

    for document_id in (
        record[
            "evidence_document_ids"
        ]
    ):

        population_document_owner_map[
            document_id
        ].append(
            record[
                "population_unit_id"
            ]
        )


population_cross_unit_document_overlap = {
    document_id:
        unit_ids

    for document_id, unit_ids
    in population_document_owner_map.items()

    if (
        len(
            set(
                unit_ids
            )
        )
        > 1
    )
}


# ============================================================
# 12) Frozen in-memory population registry
# ============================================================

EXERCISE_UNIT_POPULATION_FREEZE_V1_DRAFT = {
    "version":
        "exercise_unit_population_freeze_v1_draft",

    "status":
        "IN_MEMORY_POPULATION_FROZEN_FOR_SAMPLING_REBUILD",

    "population_count":
        len(
            exercise_unit_population_records
        ),

    "population_snapshot_sha256":
        exercise_population_snapshot_sha256,

    "legacy_selected_sample_unit_count":
        len(
            legacy_represented_population_ids
        ),

    "newly_discovered_population_unit_count":
        len(
            population_units_without_legacy_case
        ),

    "legacy_record_level_case_trace_count":
        len(
            legacy_38_to_population_mapping
        ),

    "sampling_status":
        "NOT_STARTED",

    "final_golden_membership_status":
        "NOT_STARTED",

    "legacy_case_supersession_status":
        "NOT_STARTED",

    "target_repair_status":
        "NOT_STARTED",

    "question_generation_status":
        "NOT_STARTED",

    "expected_answer_generation_status":
        "NOT_STARTED",

    "artifact_write_status":
        "NONE",
}


# ============================================================
# 13) Validation gates
# ============================================================

exact_45_population_members_gate = (
    len(
        exercise_unit_population_records
    )
    == 45
)


all_population_ids_unique_gate = (
    len(
        exercise_unit_population_by_id
    )
    == 45
)


all_structure_groups_unique_gate = (
    len(
        exercise_unit_population_by_group
    )
    == 45
)


all_population_members_eligible_gate = all(
    record[
        "eligibility_status"
    ]
    ==
    "ELIGIBLE_FOR_EXERCISE_UNIT_POPULATION"

    and

    record[
        "population_membership"
    ]
    is True

    for record
    in exercise_unit_population_records
)


population_ids_recomputable_gate = (
    len(
        population_id_recomputation_failures
    )
    == 0
)


population_fingerprints_recomputable_gate = (
    len(
        population_evidence_fingerprint_failures
    )
    == 0
)


no_cross_unit_overlap_gate = (
    len(
        population_cross_unit_document_overlap
    )
    == 0
)


exact_18_reconstructed_units_map_gate = (
    len(
        reconstructed_18_to_population_mapping
    )
    == 18

    and

    len(
        {
            row[
                "population_unit_id"
            ]
            for row
            in reconstructed_18_to_population_mapping
        }
    )
    == 18
)


exact_38_legacy_cases_trace_gate = (
    len(
        legacy_38_to_population_mapping
    )
    == 38

    and

    len(
        {
            row[
                "legacy_golden_case_id"
            ]
            for row
            in legacy_38_to_population_mapping
        }
    )
    == 38
)


legacy_18_plus_new_27_gate = (
    len(
        legacy_represented_population_ids
    )
    == 18

    and

    len(
        population_units_without_legacy_case
    )
    == 27

    and

    18
    +
    27
    ==
    45
)


work_distribution_gate = (
    exercise_population_work_counts[
        "book:adult-financial-literacy-course"
    ]
    == 31

    and

    exercise_population_work_counts[
        "book:trainer-training-financial-literacy"
    ]
    == 14
)


sampling_not_started_gate = all(
    record[
        "sampling_status"
    ]
    ==
    "NOT_STARTED"

    for record
    in exercise_unit_population_records
)


final_membership_not_started_gate = all(
    record[
        "final_golden_membership"
    ]
    is False

    for record
    in exercise_unit_population_records
)


supersession_not_started_gate = all(
    row[
        "mapping_status"
    ]
    ==
    "TRACEABILITY_PRESERVED_NO_SUPERSESSION_APPLIED"

    for row
    in legacy_38_to_population_mapping
)


target_repair_not_started_gate = (
    EXERCISE_UNIT_POPULATION_FREEZE_V1_DRAFT[
        "target_repair_status"
    ]
    ==
    "NOT_STARTED"
)


question_generation_not_started_gate = (
    EXERCISE_UNIT_POPULATION_FREEZE_V1_DRAFT[
        "question_generation_status"
    ]
    ==
    "NOT_STARTED"
)


expected_answer_not_started_gate = (
    EXERCISE_UNIT_POPULATION_FREEZE_V1_DRAFT[
        "expected_answer_generation_status"
    ]
    ==
    "NOT_STARTED"
)


no_artifact_write_gate = (
    EXERCISE_UNIT_POPULATION_FREEZE_V1_DRAFT[
        "artifact_write_status"
    ]
    ==
    "NONE"
)


cell_43_gates = {
    "exact_45_exercise_unit_population_members":
        exact_45_population_members_gate,

    "all_45_population_unit_ids_unique":
        all_population_ids_unique_gate,

    "all_45_structural_groups_unique":
        all_structure_groups_unique_gate,

    "all_population_members_have_passed_eligibility":
        all_population_members_eligible_gate,

    "all_population_ids_independently_recomputable":
        population_ids_recomputable_gate,

    "all_population_evidence_fingerprints_recomputable":
        population_fingerprints_recomputable_gate,

    "no_cross_unit_population_document_overlap":
        no_cross_unit_overlap_gate,

    "all_18_reconstructed_sample_units_map_to_frozen_population":
        exact_18_reconstructed_units_map_gate,

    "all_38_legacy_record_cases_preserve_traceability":
        exact_38_legacy_cases_trace_gate,

    "population_composition_reproduces_18_legacy_represented_plus_27_new":
        legacy_18_plus_new_27_gate,

    "population_work_distribution_reproduces_31_plus_14":
        work_distribution_gate,

    "sampling_not_started":
        sampling_not_started_gate,

    "final_golden_membership_not_started":
        final_membership_not_started_gate,

    "legacy_supersession_not_started":
        supersession_not_started_gate,

    "target_repair_not_started":
        target_repair_not_started_gate,

    "question_generation_not_started":
        question_generation_not_started_gate,

    "expected_answer_generation_not_started":
        expected_answer_not_started_gate,

    "no_artifact_write":
        no_artifact_write_gate,
}


cell_43_pass = all(
    cell_43_gates.values()
)


# ============================================================
# 14) Output
# ============================================================

print(
    "=== EXERCISE UNIT POPULATION FREEZE ==="
)

print(
    "Population members:",
    len(
        exercise_unit_population_records
    )
)

print(
    "Population snapshot SHA256:",
    exercise_population_snapshot_sha256
)

print(
    "Legacy-selected structural units represented:",
    len(
        legacy_represented_population_ids
    )
)

print(
    "New full-corpus population units:",
    len(
        population_units_without_legacy_case
    )
)

print(
    "Legacy record-level Golden Case traces:",
    len(
        legacy_38_to_population_mapping
    )
)


print(
    "\n=== POPULATION BY WORK ==="
)

for work_id, count in sorted(
    exercise_population_work_counts.items()
):

    print(
        "-",
        work_id,
        ":",
        count,
    )


print(
    "\n=== POPULATION BY TERMINAL MARKER ==="
)

for marker, count in sorted(
    exercise_population_marker_counts.items()
):

    print(
        "-",
        marker,
        ":",
        count,
    )


print(
    "\n=== DISCOVERY ORIGIN ==="
)

for origin, count in sorted(
    exercise_population_origin_counts.items()
):

    print(
        "-",
        origin,
        ":",
        count,
    )


print(
    "\n=== NEWLY DISCOVERED 27 POPULATION UNITS ==="
)

for record in (
    population_units_without_legacy_case
):

    print(
        "-",
        record[
            "population_unit_id"
        ],
        "|",
        record[
            "heading_path"
        ],
        "| records:",
        record[
            "evidence_document_count"
        ],
        "| tokens:",
        record[
            "exercise_core_token_count"
        ],
    )


print(
    "\n=== REPRODUCIBILITY ==="
)

print(
    "Population-ID recomputation failures:",
    len(
        population_id_recomputation_failures
    )
)

print(
    "Evidence-fingerprint recomputation failures:",
    len(
        population_evidence_fingerprint_failures
    )
)

print(
    "Cross-unit document overlaps:",
    len(
        population_cross_unit_document_overlap
    )
)


print(
    "\n=== CODE CELL 43 GATES ==="
)

for name, passed in (
    cell_43_gates.items()
):

    print(
        name,
        ":",
        "PASS"
        if passed
        else "FAIL"
    )


print(
    "\nCODE CELL 43 GATE:",
    "PASS"
    if cell_43_pass
    else "FAIL"
)


assert cell_43_pass, (
    "Exercise Unit population freeze failed. "
    "Do not rebuild sampling until the complete 45-unit "
    "population snapshot, stable identities, evidence "
    "fingerprints and legacy traceability are reproducible."
)
```

<h3>Sampling Rebuild Policy and Downstream Impact Audit</h3>

<p>
Cell 43 Population جدید Stratum
<code>exercise_or_self_assessment</code>
را با ۴۵ Exercise Unit پایدار Freeze کرد.
</p>

<p>
Sampling قبلی این Stratum در Cell 21 دارای Target Quota برابر با
<strong>38</strong>
بود. اکنون باید بررسی شود که آیا همین Quota پس از تغییر Identity از
Record-level به Exercise-unit-level همچنان قابل استفاده است یا خیر.
</p>

<p>
در این Cell هیچ Sample نهایی انتخاب نمی‌شود.
فقط امکان‌پذیری Policy قبلی و اثر آن بر Countهای پایین‌دستی بررسی می‌شود.
</p>

<h4>Coverage Audit</h4>

<p>
برای جلوگیری از Sampling کور، امکان پوشش هم‌زمان این Structural Dimensionها بررسی می‌شود:
</p>

<ul>
<li>هر Work موجود در Population؛</li>
<li>هر Terminal-marker signature واقعی؛</li>
<li>وجود یا عدم وجود Table در Exercise Core؛</li>
<li>هر Boundary-source type مشاهده‌شده.</li>
</ul>

<p>
این Dimensionها Structural هستند و هیچ Topic، Semantic Score یا Token Threshold جدیدی
اختراع نمی‌کنند.
</p>

<p>
یک Coverage Witness ممکن است در حافظه ساخته شود، اما این Witness
<strong>Sample نهایی نیست</strong> و هیچ Golden Membership ایجاد نمی‌کند.
</p>

<h4>دو سناریوی Count</h4>

<ul>
<li>
حفظ Quota تاریخی 38:
۳۸ Unit از Population ۴۵تایی وارد Sampling می‌شوند و ۷ Unit خارج می‌مانند.
</li>

<li>
Census کامل 45:
تمام ۴۵ Unit انتخاب می‌شوند، اما Countهای قبلی Golden Dataset تغییر خواهند کرد.
</li>
</ul>

<p>
هدف Cell تعیین این است که آیا Quota 38 بدون از دست دادن Coverage ساختاری قابل اجراست
و چه اثری روی Countهای قبلی 572 Auto / 152 Review / 724 Total Membership خواهد داشت.
</p>

<p>
هیچ Sampling، Supersession، Target Repair، Pilot Replacement یا Question Generation
در این مرحله انجام نمی‌شود.
</p>

```python
# Code Cell 44

# ============================================================
# Sampling Rebuild Policy & Downstream Impact Audit
#
# INPUT:
#   Frozen 45-unit Exercise Population from Cell 43.
#
# HISTORICAL POLICY:
#   Cell 21 target quota for exercise_or_self_assessment = 38.
#
# THIS CELL:
#   - validates quota 38 against the rebuilt population
#   - audits structural coverage feasibility
#   - compares quota-38 vs census-45 count impact
#
# IMPORTANT:
#   Coverage witness != final sample.
#
# NO sampling selection.
# NO Golden membership mutation.
# NO supersession.
# NO target repair.
# NO pilot replacement.
# NO question generation.
# NO artifact writes.
# ============================================================


# ============================================================
# 1) Historical / current audited counts
# ============================================================

CELL21_EXERCISE_TARGET_QUOTA = 38

LEGACY_EXERCISE_SELECTED_CASE_COUNT = (
    len(
        exercise_selected_cases
    )
)

LEGACY_AUTO_READY_CASE_COUNT = 572

LEGACY_REVIEW_BLOCKED_CASE_COUNT = 152

LEGACY_TOTAL_MEMBERSHIP_COUNT = 724


rebuilt_exercise_population_count = (
    len(
        exercise_unit_population_records
    )
)


quota_38_exclusion_count = (
    rebuilt_exercise_population_count
    -
    CELL21_EXERCISE_TARGET_QUOTA
)


quota_38_population_share = (
    CELL21_EXERCISE_TARGET_QUOTA
    /
    rebuilt_exercise_population_count
)


# ============================================================
# 2) Derive structural attributes for each population unit
#
# No semantic categories are invented.
# ============================================================

exercise_population_sampling_audit_rows = []


for population_record in (
    exercise_unit_population_records
):

    group_key = (
        population_record[
            "structure_group_key"
        ]
    )


    group_records = (
        book_heading_groups[
            group_key
        ]
    )


    group_record_by_document_id = {
        record[
            "document_id"
        ]:
            record

        for record
        in group_records
    }


    core_group_records = [
        group_record_by_document_id[
            document_id
        ]

        for document_id
        in population_record[
            "evidence_document_ids"
        ]
    ]


    has_table_record = any(
        record[
            "in_table"
        ]
        is True

        for record
        in core_group_records
    )


    terminal_marker_signature = tuple(
        sorted(
            population_record[
                "terminal_markers"
            ]
        )
    )


    exercise_population_sampling_audit_rows.append(
        {
            "population_unit_id":
                population_record[
                    "population_unit_id"
                ],

            "work_id":
                population_record[
                    "work_id"
                ],

            "edition_id":
                population_record[
                    "edition_id"
                ],

            "heading_path":
                population_record[
                    "heading_path"
                ],

            "terminal_marker_signature":
                terminal_marker_signature,

            "has_table_record":
                has_table_record,

            "boundary_source":
                population_record[
                    "boundary_source"
                ],

            "exercise_core_token_count":
                population_record[
                    "exercise_core_token_count"
                ],

            "evidence_document_count":
                population_record[
                    "evidence_document_count"
                ],
        }
    )


# ============================================================
# 3) Structural distributions
# ============================================================

sampling_audit_work_counts = Counter(
    row[
        "work_id"
    ]
    for row
    in exercise_population_sampling_audit_rows
)


sampling_audit_marker_signature_counts = Counter(
    row[
        "terminal_marker_signature"
    ]
    for row
    in exercise_population_sampling_audit_rows
)


sampling_audit_table_presence_counts = Counter(
    row[
        "has_table_record"
    ]
    for row
    in exercise_population_sampling_audit_rows
)


sampling_audit_boundary_source_counts = Counter(
    row[
        "boundary_source"
    ]
    for row
    in exercise_population_sampling_audit_rows
)


# ============================================================
# 4) Build structural coverage requirements
#
# Each observed category becomes one requirement.
#
# This does NOT create sample quotas by category.
# It only asks whether a quota-38 sample CAN preserve at least
# one representative of every observed structural category.
# ============================================================

sampling_coverage_requirements = set()


for work_id in (
    sampling_audit_work_counts
):

    sampling_coverage_requirements.add(
        (
            "WORK",
            work_id,
        )
    )


for marker_signature in (
    sampling_audit_marker_signature_counts
):

    sampling_coverage_requirements.add(
        (
            "TERMINAL_MARKER_SIGNATURE",
            marker_signature,
        )
    )


for has_table in (
    sampling_audit_table_presence_counts
):

    sampling_coverage_requirements.add(
        (
            "TABLE_PRESENCE",
            has_table,
        )
    )


for boundary_source in (
    sampling_audit_boundary_source_counts
):

    sampling_coverage_requirements.add(
        (
            "BOUNDARY_SOURCE",
            boundary_source,
        )
    )


# ============================================================
# 5) Requirements covered by each unit
# ============================================================

sampling_requirements_by_unit = {}


for row in (
    exercise_population_sampling_audit_rows
):

    requirements = {
        (
            "WORK",
            row[
                "work_id"
            ],
        ),

        (
            "TERMINAL_MARKER_SIGNATURE",
            row[
                "terminal_marker_signature"
            ],
        ),

        (
            "TABLE_PRESENCE",
            row[
                "has_table_record"
            ],
        ),

        (
            "BOUNDARY_SOURCE",
            row[
                "boundary_source"
            ],
        ),
    }


    sampling_requirements_by_unit[
        row[
            "population_unit_id"
        ]
    ] = (
        requirements
    )


# ============================================================
# 6) Deterministic tie key for FEASIBILITY WITNESS ONLY
#
# This is NOT the final sampling selector.
# ============================================================

def sampling_audit_tie_key(
    population_unit_id,
):

    return hashlib.sha256(
        stable_json_key(
            {
                "purpose":
                    "CELL44_COVERAGE_FEASIBILITY_WITNESS_ONLY",

                "population_snapshot_sha256":
                    exercise_population_snapshot_sha256,

                "population_unit_id":
                    population_unit_id,
            }
        ).encode(
            "utf-8"
        )
    ).hexdigest()


# ============================================================
# 7) Build a deterministic set-cover witness
#
# Purpose:
# prove whether all observed structural requirements can fit
# inside quota 38.
#
# NOT final sampling.
# ============================================================

uncovered_requirements = set(
    sampling_coverage_requirements
)


coverage_witness_unit_ids = []


while uncovered_requirements:

    candidate_rows = []


    for row in (
        exercise_population_sampling_audit_rows
    ):

        population_unit_id = (
            row[
                "population_unit_id"
            ]
        )


        if (
            population_unit_id
            in
            coverage_witness_unit_ids
        ):

            continue


        newly_covered = (
            sampling_requirements_by_unit[
                population_unit_id
            ]
            &
            uncovered_requirements
        )


        if not newly_covered:
            continue


        candidate_rows.append(
            {
                "population_unit_id":
                    population_unit_id,

                "new_coverage_count":
                    len(
                        newly_covered
                    ),

                "newly_covered":
                    newly_covered,

                "tie_key":
                    sampling_audit_tie_key(
                        population_unit_id
                    ),
            }
        )


    if not candidate_rows:

        break


    selected_witness_row = sorted(
        candidate_rows,
        key=lambda item: (
            -item[
                "new_coverage_count"
            ],
            item[
                "tie_key"
            ],
        ),
    )[0]


    coverage_witness_unit_ids.append(
        selected_witness_row[
            "population_unit_id"
        ]
    )


    uncovered_requirements -= (
        selected_witness_row[
            "newly_covered"
        ]
    )


coverage_witness_size = (
    len(
        coverage_witness_unit_ids
    )
)


coverage_witness_remaining_capacity_at_quota_38 = (
    CELL21_EXERCISE_TARGET_QUOTA
    -
    coverage_witness_size
)


# ============================================================
# 8) Independently recompute coverage obtained by witness
# ============================================================

coverage_witness_recomputed_requirements = set()


for population_unit_id in (
    coverage_witness_unit_ids
):

    coverage_witness_recomputed_requirements |= (
        sampling_requirements_by_unit[
            population_unit_id
        ]
    )


coverage_witness_missing_requirements = (
    sampling_coverage_requirements
    -
    coverage_witness_recomputed_requirements
)


quota_38_structural_coverage_feasible = (
    coverage_witness_size
    <=
    CELL21_EXERCISE_TARGET_QUOTA

    and

    len(
        coverage_witness_missing_requirements
    )
    == 0

    and

    rebuilt_exercise_population_count
    >=
    CELL21_EXERCISE_TARGET_QUOTA
)


# ============================================================
# 9) Count impact scenarios
#
# No mutation is performed.
# ============================================================

# ------------------------------------------------------------
# Scenario A:
# Preserve Cell 21 quota = 38
# ------------------------------------------------------------

quota38_projected_auto_count = (
    LEGACY_AUTO_READY_CASE_COUNT
    -
    LEGACY_EXERCISE_SELECTED_CASE_COUNT
    +
    CELL21_EXERCISE_TARGET_QUOTA
)


quota38_projected_review_count = (
    LEGACY_REVIEW_BLOCKED_CASE_COUNT
)


quota38_projected_total_membership_count = (
    LEGACY_TOTAL_MEMBERSHIP_COUNT
    -
    LEGACY_EXERCISE_SELECTED_CASE_COUNT
    +
    CELL21_EXERCISE_TARGET_QUOTA
)


# ------------------------------------------------------------
# Scenario B:
# Census all 45 exercise units
# ------------------------------------------------------------

census45_projected_auto_count = (
    LEGACY_AUTO_READY_CASE_COUNT
    -
    LEGACY_EXERCISE_SELECTED_CASE_COUNT
    +
    rebuilt_exercise_population_count
)


census45_projected_review_count = (
    LEGACY_REVIEW_BLOCKED_CASE_COUNT
)


census45_projected_total_membership_count = (
    LEGACY_TOTAL_MEMBERSHIP_COUNT
    -
    LEGACY_EXERCISE_SELECTED_CASE_COUNT
    +
    rebuilt_exercise_population_count
)


# ============================================================
# 10) Pilot / downstream dependency impact
#
# Descriptive only.
# ============================================================

sampling_rebuild_downstream_impact = {
    "exercise_case_identity":
        "MUST_CHANGE_FROM_RECORD_LEVEL_TO_EXERCISE_UNIT_LEVEL",

    "exercise_target_evidence":
        "MUST_BE_REBUILT_AFTER_NEW_SAMPLE",

    "exercise_pilot_case":
        "MUST_BE_RESELECTED_FROM_REBUILT_EXERCISE_SAMPLE",

    "overall_pilot_stratum_count":
        "CAN_REMAIN_ONE_PILOT_FOR_THIS_READY_STRATUM",

    "other_strata_sampling":
        "NO_CHANGE_REQUIRED_BY_THIS_REPAIR",

    "review_queue_count":
        "NO_CHANGE_REQUIRED_BY_THIS_REPAIR",

    "question_generation":
        "STILL_NOT_STARTED",
}


# ============================================================
# 11) Sampling policy audit classification
# ============================================================

if (
    quota_38_structural_coverage_feasible

    and

    quota38_projected_auto_count
    ==
    LEGACY_AUTO_READY_CASE_COUNT

    and

    quota38_projected_total_membership_count
    ==
    LEGACY_TOTAL_MEMBERSHIP_COUNT
):

    sampling_rebuild_policy_classification = (
        "PRESERVE_CELL21_QUOTA_38_IS_FEASIBLE_AND_COUNT_STABLE"
    )

else:

    sampling_rebuild_policy_classification = (
        "CELL21_QUOTA_38_REQUIRES_POLICY_REVIEW"
    )


# ============================================================
# 12) Audit registry
# ============================================================

EXERCISE_SAMPLING_REBUILD_AUDIT_V1_DRAFT = {
    "version":
        "exercise_sampling_rebuild_audit_v1_draft",

    "status":
        "POLICY_AUDITED_SELECTION_NOT_STARTED",

    "population_snapshot_sha256":
        exercise_population_snapshot_sha256,

    "population_count":
        rebuilt_exercise_population_count,

    "historical_cell21_target_quota":
        CELL21_EXERCISE_TARGET_QUOTA,

    "legacy_selected_exercise_case_count":
        LEGACY_EXERCISE_SELECTED_CASE_COUNT,

    "quota_38_exclusion_count":
        quota_38_exclusion_count,

    "quota_38_population_share":
        quota_38_population_share,

    "structural_coverage_requirement_count":
        len(
            sampling_coverage_requirements
        ),

    "coverage_witness_size":
        coverage_witness_size,

    "coverage_witness_only":
        True,

    "quota_38_structural_coverage_feasible":
        quota_38_structural_coverage_feasible,

    "policy_classification":
        sampling_rebuild_policy_classification,

    "sampling_selection_status":
        "NOT_STARTED",

    "legacy_supersession_status":
        "NOT_STARTED",

    "target_repair_status":
        "NOT_STARTED",

    "pilot_replacement_status":
        "NOT_STARTED",

    "question_generation_status":
        "NOT_STARTED",

    "expected_answer_generation_status":
        "NOT_STARTED",

    "artifact_write_status":
        "NONE",
}


# ============================================================
# 13) Mutation guardrails
# ============================================================

population_sampling_status_unchanged = all(
    record[
        "sampling_status"
    ]
    ==
    "NOT_STARTED"

    for record
    in exercise_unit_population_records
)


population_final_membership_unchanged = all(
    record[
        "final_golden_membership"
    ]
    is False

    for record
    in exercise_unit_population_records
)


# ============================================================
# 14) Validation gates
# ============================================================

historical_quota_reproduced_gate = (
    CELL21_EXERCISE_TARGET_QUOTA
    == 38

    and

    LEGACY_EXERCISE_SELECTED_CASE_COUNT
    == 38
)


population_45_gate = (
    rebuilt_exercise_population_count
    == 45
)


quota_exclusion_math_gate = (
    quota_38_exclusion_count
    == 7
)


all_structural_requirements_have_candidates_gate = all(
    any(
        requirement
        in
        sampling_requirements_by_unit[
            row[
                "population_unit_id"
            ]
        ]

        for row
        in exercise_population_sampling_audit_rows
    )

    for requirement
    in sampling_coverage_requirements
)


coverage_witness_complete_gate = (
    len(
        coverage_witness_missing_requirements
    )
    == 0
)


coverage_witness_fits_quota_gate = (
    coverage_witness_size
    <=
    CELL21_EXERCISE_TARGET_QUOTA
)


quota38_coverage_feasible_gate = (
    quota_38_structural_coverage_feasible
    is True
)


quota38_preserves_auto_count_gate = (
    quota38_projected_auto_count
    ==
    572
)


quota38_preserves_review_count_gate = (
    quota38_projected_review_count
    ==
    152
)


quota38_preserves_total_membership_count_gate = (
    quota38_projected_total_membership_count
    ==
    724
)


census45_count_math_gate = (
    census45_projected_auto_count
    ==
    579

    and

    census45_projected_review_count
    ==
    152

    and

    census45_projected_total_membership_count
    ==
    731
)


policy_classification_resolved_gate = (
    sampling_rebuild_policy_classification
    in {
        "PRESERVE_CELL21_QUOTA_38_IS_FEASIBLE_AND_COUNT_STABLE",
        "CELL21_QUOTA_38_REQUIRES_POLICY_REVIEW",
    }
)


coverage_witness_not_sample_gate = (
    EXERCISE_SAMPLING_REBUILD_AUDIT_V1_DRAFT[
        "coverage_witness_only"
    ]
    is True

    and

    EXERCISE_SAMPLING_REBUILD_AUDIT_V1_DRAFT[
        "sampling_selection_status"
    ]
    ==
    "NOT_STARTED"
)


sampling_not_started_gate = (
    population_sampling_status_unchanged
)


final_membership_not_started_gate = (
    population_final_membership_unchanged
)


supersession_not_started_gate = (
    EXERCISE_SAMPLING_REBUILD_AUDIT_V1_DRAFT[
        "legacy_supersession_status"
    ]
    ==
    "NOT_STARTED"
)


target_repair_not_started_gate = (
    EXERCISE_SAMPLING_REBUILD_AUDIT_V1_DRAFT[
        "target_repair_status"
    ]
    ==
    "NOT_STARTED"
)


pilot_replacement_not_started_gate = (
    EXERCISE_SAMPLING_REBUILD_AUDIT_V1_DRAFT[
        "pilot_replacement_status"
    ]
    ==
    "NOT_STARTED"
)


question_generation_not_started_gate = (
    EXERCISE_SAMPLING_REBUILD_AUDIT_V1_DRAFT[
        "question_generation_status"
    ]
    ==
    "NOT_STARTED"
)


expected_answer_not_started_gate = (
    EXERCISE_SAMPLING_REBUILD_AUDIT_V1_DRAFT[
        "expected_answer_generation_status"
    ]
    ==
    "NOT_STARTED"
)


no_artifact_write_gate = (
    EXERCISE_SAMPLING_REBUILD_AUDIT_V1_DRAFT[
        "artifact_write_status"
    ]
    ==
    "NONE"
)


cell_44_gates = {
    "cell21_exercise_quota_38_reproduced":
        historical_quota_reproduced_gate,

    "rebuilt_exercise_population_is_45":
        population_45_gate,

    "quota_38_requires_exactly_7_exclusions":
        quota_exclusion_math_gate,

    "all_observed_structural_coverage_requirements_have_candidates":
        all_structural_requirements_have_candidates_gate,

    "coverage_feasibility_witness_covers_all_requirements":
        coverage_witness_complete_gate,

    "coverage_feasibility_witness_fits_inside_quota_38":
        coverage_witness_fits_quota_gate,

    "quota_38_structural_coverage_is_feasible":
        quota38_coverage_feasible_gate,

    "quota_38_preserves_legacy_auto_count_572":
        quota38_preserves_auto_count_gate,

    "quota_38_preserves_review_count_152":
        quota38_preserves_review_count_gate,

    "quota_38_preserves_total_membership_count_724":
        quota38_preserves_total_membership_count_gate,

    "census_45_count_impact_recomputes_to_579_152_731":
        census45_count_math_gate,

    "sampling_policy_classification_resolved":
        policy_classification_resolved_gate,

    "coverage_witness_is_not_final_sample":
        coverage_witness_not_sample_gate,

    "actual_sampling_not_started":
        sampling_not_started_gate,

    "final_golden_membership_not_started":
        final_membership_not_started_gate,

    "legacy_supersession_not_started":
        supersession_not_started_gate,

    "target_repair_not_started":
        target_repair_not_started_gate,

    "pilot_replacement_not_started":
        pilot_replacement_not_started_gate,

    "question_generation_not_started":
        question_generation_not_started_gate,

    "expected_answer_generation_not_started":
        expected_answer_not_started_gate,

    "no_artifact_write":
        no_artifact_write_gate,
}


cell_44_pass = all(
    cell_44_gates.values()
)


# ============================================================
# 15) Output
# ============================================================

print(
    "=== EXERCISE SAMPLING REBUILD POLICY AUDIT ==="
)

print(
    "Frozen population:",
    rebuilt_exercise_population_count
)

print(
    "Cell 21 historical target quota:",
    CELL21_EXERCISE_TARGET_QUOTA
)

print(
    "Legacy selected exercise cases:",
    LEGACY_EXERCISE_SELECTED_CASE_COUNT
)

print(
    "Units excluded if quota 38 is preserved:",
    quota_38_exclusion_count
)

print(
    "Quota-38 population share:",
    quota_38_population_share
)


print(
    "\n=== STRUCTURAL COVERAGE DISTRIBUTIONS ==="
)

print(
    "Work:",
    dict(
        sampling_audit_work_counts
    )
)

print(
    "Terminal-marker signatures:"
)

for signature, count in sorted(
    sampling_audit_marker_signature_counts.items(),
    key=lambda item:
        str(
            item[
                0
            ]
        ),
):

    print(
        "-",
        signature,
        ":",
        count,
    )


print(
    "Table presence:",
    dict(
        sampling_audit_table_presence_counts
    )
)

print(
    "Boundary source:",
    dict(
        sampling_audit_boundary_source_counts
    )
)


print(
    "\n=== COVERAGE FEASIBILITY ==="
)

print(
    "Structural requirements:",
    len(
        sampling_coverage_requirements
    )
)

print(
    "Coverage witness size:",
    coverage_witness_size
)

print(
    "Remaining capacity inside quota 38:",
    coverage_witness_remaining_capacity_at_quota_38
)

print(
    "Missing requirements:",
    len(
        coverage_witness_missing_requirements
    )
)

print(
    "Quota 38 structurally feasible:",
    quota_38_structural_coverage_feasible
)


print(
    "\nCoverage witness units (NOT FINAL SAMPLE):"
)

for population_unit_id in (
    coverage_witness_unit_ids
):

    row = next(
        row
        for row
        in exercise_population_sampling_audit_rows
        if (
            row[
                "population_unit_id"
            ]
            ==
            population_unit_id
        )
    )

    print(
        "-",
        population_unit_id,
        "| work:",
        row[
            "work_id"
        ],
        "| markers:",
        row[
            "terminal_marker_signature"
        ],
        "| table:",
        row[
            "has_table_record"
        ],
        "| boundary:",
        row[
            "boundary_source"
        ],
        "| heading:",
        row[
            "heading_path"
        ],
    )


print(
    "\n=== DOWNSTREAM COUNT IMPACT ==="
)

print(
    "Scenario A — preserve quota 38:"
)

print(
    {
        "auto":
            quota38_projected_auto_count,

        "review":
            quota38_projected_review_count,

        "total_memberships":
            quota38_projected_total_membership_count,

        "exercise_sample":
            CELL21_EXERCISE_TARGET_QUOTA,

        "exercise_population_excluded":
            quota_38_exclusion_count,
    }
)


print(
    "Scenario B — census 45:"
)

print(
    {
        "auto":
            census45_projected_auto_count,

        "review":
            census45_projected_review_count,

        "total_memberships":
            census45_projected_total_membership_count,

        "exercise_sample":
            rebuilt_exercise_population_count,

        "exercise_population_excluded":
            0,
    }
)


print(
    "\n=== DOWNSTREAM DEPENDENCY IMPACT ==="
)

for name, status in (
    sampling_rebuild_downstream_impact.items()
):

    print(
        "-",
        name,
        ":",
        status,
    )


print(
    "\nPolicy classification:",
    sampling_rebuild_policy_classification
)


print(
    "\n=== CODE CELL 44 GATES ==="
)

for name, passed in (
    cell_44_gates.items()
):

    print(
        name,
        ":",
        "PASS"
        if passed
        else "FAIL"
    )


print(
    "\nCODE CELL 44 GATE:",
    "PASS"
    if cell_44_pass
    else "FAIL"
)


assert cell_44_pass, (
    "Exercise sampling rebuild policy audit failed. "
    "Do not select the rebuilt 38-unit exercise sample until "
    "quota feasibility, structural coverage and downstream "
    "count impact are fully validated."
)
```

<div style="direction: rtl; white-space: normal; line-height: 1.7;">

<h3>Deterministic Rebuilt Exercise Sampling</h3>

<p>
Cell 44 تأیید کرد که Quota تاریخی 38 روی Population جدید 45تایی
قابل حفظ است و Countهای کلی Golden Dataset را تغییر نمی‌دهد.
</p>

<p>
در این Cell انتخاب واقعی 38 Exercise Unit انجام می‌شود.
Selection باید deterministic، قابل بازتولید و Coverage-safe باشد.
</p>

<h4>Work-Proportional Exclusion</h4>

<p>
به‌جای تعیین سهم دلخواه، تعداد 7 حذف بر اساس نسبت واقعی Population بین دو Work
با روش Largest Remainder محاسبه می‌شود:
</p>

<ul>
<li>Adult Financial Literacy Course: Population = 31</li>
<li>Trainer Training Financial Literacy: Population = 14</li>
<li>Total exclusions = 7</li>
</ul>

<p>
این محاسبه باید به 5 حذف از Work اول و 2 حذف از Work دوم برسد؛
بنابراین Sample نهایی شامل 26 + 12 = 38 Unit خواهد بود.
</p>

<h4>Coverage Guardrail</h4>

<p>
یک Unit فقط زمانی می‌تواند حذف شود که بعد از حذف آن هنوز حداقل یک نماینده برای تمام
Structural Requirementهای Cell 44 باقی بماند:
</p>

<ul>
<li>Work</li>
<li>Terminal-marker signature</li>
<li>Table presence</li>
<li>Boundary source</li>
</ul>

<p>
ترتیب بررسی حذف‌ها با Stable Hash وابسته به Population Snapshot تعیین می‌شود.
هیچ Random Sampling انجام نمی‌شود.
</p>

<p>
خروجی این Cell فقط Sample Membership جدید است.
۳۸ Golden Case قدیمی هنوز Supersede نمی‌شوند،
Target Evidence بازسازی نمی‌شود،
Pilot جدید انتخاب نمی‌شود و Question Generation شروع نمی‌شود.
</p>

</div>

```python
# Code Cell 45

# ============================================================
# Deterministic Rebuilt Exercise Sampling
#
# Population:
#   45 frozen Exercise Units
#
# Target:
#   38 selected
#   7 excluded
#
# Exclusion allocation:
#   proportionally derived by Largest Remainder across Works.
#
# Coverage guardrail:
#   no exclusion may remove the final remaining representative
#   of any structural requirement defined in Cell 44.
#
# NO legacy supersession.
# NO Target Evidence rebuild.
# NO pilot replacement.
# NO question generation.
# NO artifact writes.
# ============================================================


# ============================================================
# 1) Constants
# ============================================================

EXERCISE_SAMPLE_TARGET_SIZE = 38

EXERCISE_POPULATION_SIZE = (
    len(
        exercise_unit_population_records
    )
)

EXERCISE_EXCLUSION_TARGET_SIZE = (
    EXERCISE_POPULATION_SIZE
    -
    EXERCISE_SAMPLE_TARGET_SIZE
)


assert (
    EXERCISE_POPULATION_SIZE
    == 45
)

assert (
    EXERCISE_SAMPLE_TARGET_SIZE
    == 38
)

assert (
    EXERCISE_EXCLUSION_TARGET_SIZE
    == 7
)


# ============================================================
# 2) Work counts
# ============================================================

exercise_population_counts_by_work = Counter(
    record[
        "work_id"
    ]
    for record
    in exercise_unit_population_records
)


# ============================================================
# 3) Largest-Remainder allocation of the 7 exclusions
#
# Formula:
#
# exact_exclusions(work)
#   = population(work) / total_population * 7
#
# Start with floor values, then allocate remaining exclusions
# to largest fractional remainders.
# ============================================================

work_exclusion_allocation_rows = []


for work_id, population_count in sorted(
    exercise_population_counts_by_work.items()
):

    exact_exclusion_count = (
        population_count
        *
        EXERCISE_EXCLUSION_TARGET_SIZE
        /
        EXERCISE_POPULATION_SIZE
    )


    floor_exclusion_count = int(
        exact_exclusion_count
    )


    fractional_remainder = (
        exact_exclusion_count
        -
        floor_exclusion_count
    )


    work_exclusion_allocation_rows.append(
        {
            "work_id":
                work_id,

            "population_count":
                population_count,

            "exact_exclusion_count":
                exact_exclusion_count,

            "floor_exclusion_count":
                floor_exclusion_count,

            "fractional_remainder":
                fractional_remainder,

            "final_exclusion_quota":
                floor_exclusion_count,
        }
    )


allocated_floor_total = sum(
    row[
        "floor_exclusion_count"
    ]
    for row
    in work_exclusion_allocation_rows
)


remaining_exclusions_to_allocate = (
    EXERCISE_EXCLUSION_TARGET_SIZE
    -
    allocated_floor_total
)


largest_remainder_order = sorted(
    work_exclusion_allocation_rows,
    key=lambda row: (
        -row[
            "fractional_remainder"
        ],
        row[
            "work_id"
        ],
    ),
)


for index in range(
    remaining_exclusions_to_allocate
):

    largest_remainder_order[
        index
    ][
        "final_exclusion_quota"
    ] += 1


work_exclusion_quota = {
    row[
        "work_id"
    ]:
        row[
            "final_exclusion_quota"
        ]

    for row
    in work_exclusion_allocation_rows
}


work_selection_quota = {
    work_id:
        (
            exercise_population_counts_by_work[
                work_id
            ]
            -
            exclusion_quota
        )

    for work_id, exclusion_quota
    in work_exclusion_quota.items()
}


# ============================================================
# 4) Stable deterministic exclusion hash
# ============================================================

def exercise_sampling_exclusion_hash(
    population_unit_id,
):

    payload = {
        "purpose":
            "NB12_EXERCISE_UNIT_SAMPLE_V1_EXCLUSION_ORDER",

        "population_snapshot_sha256":
            exercise_population_snapshot_sha256,

        "target_sample_size":
            EXERCISE_SAMPLE_TARGET_SIZE,

        "population_unit_id":
            population_unit_id,
    }


    return hashlib.sha256(
        stable_json_key(
            payload
        ).encode(
            "utf-8"
        )
    ).hexdigest()


# ============================================================
# 5) Sampling audit lookup
# ============================================================

sampling_audit_row_by_unit_id = {
    row[
        "population_unit_id"
    ]:
        row

    for row
    in exercise_population_sampling_audit_rows
}


assert (
    len(
        sampling_audit_row_by_unit_id
    )
    == 45
)


# ============================================================
# 6) Deterministic coverage-safe exclusion function
#
# A unit can be excluded only if:
#
# 1. its Work still needs an exclusion;
# 2. removing it leaves at least one representative for each
#    structural requirement it currently covers.
#
# The initial remaining set is the complete population.
# ============================================================

def build_deterministic_exercise_exclusion_set():

    remaining_unit_ids = {
        record[
            "population_unit_id"
        ]
        for record
        in exercise_unit_population_records
    }


    remaining_requirement_counts = Counter()


    for population_unit_id in (
        remaining_unit_ids
    ):

        for requirement in (
            sampling_requirements_by_unit[
                population_unit_id
            ]
        ):

            remaining_requirement_counts[
                requirement
            ] += 1


    remaining_work_exclusion_quota = dict(
        work_exclusion_quota
    )


    excluded_rows = []

    skipped_for_coverage_rows = []


    ordered_candidates = sorted(
        exercise_unit_population_records,
        key=lambda record: (
            exercise_sampling_exclusion_hash(
                record[
                    "population_unit_id"
                ]
            ),
            record[
                "population_unit_id"
            ],
        ),
    )


    for record in (
        ordered_candidates
    ):

        if (
            len(
                excluded_rows
            )
            >=
            EXERCISE_EXCLUSION_TARGET_SIZE
        ):

            break


        population_unit_id = (
            record[
                "population_unit_id"
            ]
        )


        work_id = (
            record[
                "work_id"
            ]
        )


        # ----------------------------------------------------
        # Work quota already satisfied -> not considered.
        # ----------------------------------------------------

        if (
            remaining_work_exclusion_quota[
                work_id
            ]
            <= 0
        ):

            continue


        covered_requirements = (
            sampling_requirements_by_unit[
                population_unit_id
            ]
        )


        blocking_requirements = [
            requirement

            for requirement
            in covered_requirements

            if (
                remaining_requirement_counts[
                    requirement
                ]
                <= 1
            )
        ]


        # ----------------------------------------------------
        # Coverage guardrail
        # ----------------------------------------------------

        if (
            blocking_requirements
        ):

            skipped_for_coverage_rows.append(
                {
                    "population_unit_id":
                        population_unit_id,

                    "work_id":
                        work_id,

                    "heading_path":
                        record[
                            "heading_path"
                        ],

                    "blocking_requirements":
                        sorted(
                            blocking_requirements,
                            key=str,
                        ),

                    "exclusion_hash":
                        exercise_sampling_exclusion_hash(
                            population_unit_id
                        ),
                }
            )

            continue


        # ----------------------------------------------------
        # Exclude unit
        # ----------------------------------------------------

        requirement_counts_before = {
            requirement:
                remaining_requirement_counts[
                    requirement
                ]

            for requirement
            in covered_requirements
        }


        remaining_unit_ids.remove(
            population_unit_id
        )


        for requirement in (
            covered_requirements
        ):

            remaining_requirement_counts[
                requirement
            ] -= 1


        remaining_work_exclusion_quota[
            work_id
        ] -= 1


        requirement_counts_after = {
            requirement:
                remaining_requirement_counts[
                    requirement
                ]

            for requirement
            in covered_requirements
        }


        excluded_rows.append(
            {
                "population_unit_id":
                    population_unit_id,

                "work_id":
                    work_id,

                "heading_path":
                    record[
                        "heading_path"
                    ],

                "terminal_markers":
                    record[
                        "terminal_markers"
                    ],

                "boundary_source":
                    record[
                        "boundary_source"
                    ],

                "has_table_record":
                    sampling_audit_row_by_unit_id[
                        population_unit_id
                    ][
                        "has_table_record"
                    ],

                "exercise_core_token_count":
                    record[
                        "exercise_core_token_count"
                    ],

                "evidence_document_count":
                    record[
                        "evidence_document_count"
                    ],

                "exclusion_hash":
                    exercise_sampling_exclusion_hash(
                        population_unit_id
                    ),

                "requirement_counts_before":
                    requirement_counts_before,

                "requirement_counts_after":
                    requirement_counts_after,

                "selection_reason":
                    (
                        "DETERMINISTIC_WORK_PROPORTIONAL_"
                        "COVERAGE_SAFE_EXCLUSION"
                    ),
            }
        )


    return {
        "remaining_unit_ids":
            remaining_unit_ids,

        "excluded_rows":
            excluded_rows,

        "skipped_for_coverage_rows":
            skipped_for_coverage_rows,

        "remaining_work_exclusion_quota":
            remaining_work_exclusion_quota,

        "remaining_requirement_counts":
            remaining_requirement_counts,
    }


# ============================================================
# 7) Execute deterministic selection
# ============================================================

exercise_sampling_result = (
    build_deterministic_exercise_exclusion_set()
)


selected_exercise_population_unit_ids = (
    exercise_sampling_result[
        "remaining_unit_ids"
    ]
)


excluded_exercise_unit_rows = (
    exercise_sampling_result[
        "excluded_rows"
    ]
)


coverage_protected_skip_rows = (
    exercise_sampling_result[
        "skipped_for_coverage_rows"
    ]
)


remaining_work_exclusion_quota = (
    exercise_sampling_result[
        "remaining_work_exclusion_quota"
    ]
)


# ============================================================
# 8) Build selected Sample records
# ============================================================

selected_exercise_unit_sample_records = [
    {
        "population_unit_id":
            record[
                "population_unit_id"
            ],

        "work_id":
            record[
                "work_id"
            ],

        "edition_id":
            record[
                "edition_id"
            ],

        "heading_path":
            record[
                "heading_path"
            ],

        "terminal_markers":
            record[
                "terminal_markers"
            ],

        "boundary_source":
            record[
                "boundary_source"
            ],

        "evidence_document_ids":
            record[
                "evidence_document_ids"
            ],

        "evidence_document_count":
            record[
                "evidence_document_count"
            ],

        "exercise_core_token_count":
            record[
                "exercise_core_token_count"
            ],

        "exercise_core_evidence_sha256":
            record[
                "exercise_core_evidence_sha256"
            ],

        "sample_membership":
            True,

        "sample_status":
            "SELECTED_FOR_REBUILT_EXERCISE_SAMPLE",

        "final_golden_membership":
            False,

        "legacy_supersession_status":
            "NOT_APPLIED",

        "target_evidence_status":
            "NOT_STARTED",
    }

    for record
    in exercise_unit_population_records

    if (
        record[
            "population_unit_id"
        ]
        in
        selected_exercise_population_unit_ids
    )
]


selected_exercise_unit_sample_records = sorted(
    selected_exercise_unit_sample_records,
    key=lambda record:
        record[
            "population_unit_id"
        ],
)


# ============================================================
# 9) Build explicit excluded set
# ============================================================

excluded_exercise_population_unit_ids = {
    row[
        "population_unit_id"
    ]
    for row
    in excluded_exercise_unit_rows
}


# ============================================================
# 10) Independent Work-count validation
# ============================================================

selected_sample_work_counts = Counter(
    record[
        "work_id"
    ]
    for record
    in selected_exercise_unit_sample_records
)


excluded_sample_work_counts = Counter(
    row[
        "work_id"
    ]
    for row
    in excluded_exercise_unit_rows
)


# ============================================================
# 11) Independently recompute all structural coverage
# ============================================================

selected_sample_requirement_counts = Counter()


for population_unit_id in (
    selected_exercise_population_unit_ids
):

    for requirement in (
        sampling_requirements_by_unit[
            population_unit_id
        ]
    ):

        selected_sample_requirement_counts[
            requirement
        ] += 1


selected_sample_missing_requirements = sorted(
    [
        requirement

        for requirement
        in sampling_coverage_requirements

        if (
            selected_sample_requirement_counts[
                requirement
            ]
            <= 0
        )
    ],
    key=str,
)


# ============================================================
# 12) Selected-sample structural distributions
# ============================================================

selected_sample_marker_signature_counts = Counter(
    sampling_audit_row_by_unit_id[
        population_unit_id
    ][
        "terminal_marker_signature"
    ]

    for population_unit_id
    in selected_exercise_population_unit_ids
)


selected_sample_table_presence_counts = Counter(
    sampling_audit_row_by_unit_id[
        population_unit_id
    ][
        "has_table_record"
    ]

    for population_unit_id
    in selected_exercise_population_unit_ids
)


selected_sample_boundary_source_counts = Counter(
    sampling_audit_row_by_unit_id[
        population_unit_id
    ][
        "boundary_source"
    ]

    for population_unit_id
    in selected_exercise_population_unit_ids
)


# ============================================================
# 13) Independent deterministic recomputation
# ============================================================

exercise_sampling_recomputed = (
    build_deterministic_exercise_exclusion_set()
)


recomputed_selected_ids = (
    exercise_sampling_recomputed[
        "remaining_unit_ids"
    ]
)


recomputed_excluded_ids = {
    row[
        "population_unit_id"
    ]
    for row
    in exercise_sampling_recomputed[
        "excluded_rows"
    ]
}


# ============================================================
# 14) Sample snapshot SHA256
# ============================================================

exercise_sample_snapshot_payload = [
    {
        "population_unit_id":
            record[
                "population_unit_id"
            ],

        "work_id":
            record[
                "work_id"
            ],

        "edition_id":
            record[
                "edition_id"
            ],

        "heading_path":
            record[
                "heading_path"
            ],

        "boundary_source":
            record[
                "boundary_source"
            ],

        "evidence_document_ids":
            record[
                "evidence_document_ids"
            ],

        "exercise_core_evidence_sha256":
            record[
                "exercise_core_evidence_sha256"
            ],
    }

    for record
    in selected_exercise_unit_sample_records
]


exercise_sample_snapshot_sha256 = hashlib.sha256(
    stable_json_key(
        exercise_sample_snapshot_payload
    ).encode(
        "utf-8"
    )
).hexdigest()


# ============================================================
# 15) Legacy representation audit
#
# Descriptive only.
#
# We do NOT require every one of the 18 legacy-represented
# structural units to survive the rebuilt sample.
# ============================================================

selected_sample_legacy_represented_ids = (
    selected_exercise_population_unit_ids
    &
    legacy_represented_population_ids
)


selected_sample_new_discovery_ids = (
    selected_exercise_population_unit_ids
    -
    legacy_represented_population_ids
)


excluded_legacy_represented_ids = (
    excluded_exercise_population_unit_ids
    &
    legacy_represented_population_ids
)


excluded_new_discovery_ids = (
    excluded_exercise_population_unit_ids
    -
    legacy_represented_population_ids
)


# ============================================================
# 16) Sampling Registry
# ============================================================

EXERCISE_UNIT_REBUILT_SAMPLE_V1_DRAFT = {
    "version":
        "exercise_unit_rebuilt_sample_v1_draft",

    "status":
        "DETERMINISTIC_38_OF_45_SELECTED_IN_MEMORY",

    "population_snapshot_sha256":
        exercise_population_snapshot_sha256,

    "sample_snapshot_sha256":
        exercise_sample_snapshot_sha256,

    "population_count":
        EXERCISE_POPULATION_SIZE,

    "sample_count":
        len(
            selected_exercise_unit_sample_records
        ),

    "excluded_count":
        len(
            excluded_exercise_unit_rows
        ),

    "work_exclusion_quota":
        dict(
            work_exclusion_quota
        ),

    "work_selection_quota":
        dict(
            work_selection_quota
        ),

    "coverage_requirement_count":
        len(
            sampling_coverage_requirements
        ),

    "coverage_missing_requirement_count":
        len(
            selected_sample_missing_requirements
        ),

    "selection_policy": (
        "LARGEST_REMAINDER_WORK_PROPORTIONAL_"
        "DETERMINISTIC_COVERAGE_SAFE_EXCLUSION"
    ),

    "selection_applied":
        True,

    "final_golden_membership_status":
        "NOT_STARTED",

    "legacy_case_supersession_status":
        "NOT_STARTED",

    "target_repair_status":
        "NOT_STARTED",

    "pilot_replacement_status":
        "NOT_STARTED",

    "question_generation_status":
        "NOT_STARTED",

    "expected_answer_generation_status":
        "NOT_STARTED",

    "artifact_write_status":
        "NONE",
}


# ============================================================
# 17) Validation gates
# ============================================================

largest_remainder_allocation_gate = (
    work_exclusion_quota[
        "book:adult-financial-literacy-course"
    ]
    == 5

    and

    work_exclusion_quota[
        "book:trainer-training-financial-literacy"
    ]
    == 2
)


selection_work_quota_gate = (
    work_selection_quota[
        "book:adult-financial-literacy-course"
    ]
    == 26

    and

    work_selection_quota[
        "book:trainer-training-financial-literacy"
    ]
    == 12
)


exact_7_excluded_gate = (
    len(
        excluded_exercise_unit_rows
    )
    == 7
)


exact_38_selected_gate = (
    len(
        selected_exercise_unit_sample_records
    )
    == 38
)


selected_excluded_partition_gate = (
    selected_exercise_population_unit_ids
    .isdisjoint(
        excluded_exercise_population_unit_ids
    )

    and

    (
        selected_exercise_population_unit_ids
        |
        excluded_exercise_population_unit_ids
    )
    ==
    {
        record[
            "population_unit_id"
        ]
        for record
        in exercise_unit_population_records
    }
)


all_work_exclusion_quotas_satisfied_gate = all(
    value
    == 0

    for value
    in remaining_work_exclusion_quota.values()
)


excluded_work_counts_match_quota_gate = all(
    excluded_sample_work_counts[
        work_id
    ]
    ==
    exclusion_quota

    for work_id, exclusion_quota
    in work_exclusion_quota.items()
)


selected_work_counts_match_quota_gate = all(
    selected_sample_work_counts[
        work_id
    ]
    ==
    selection_quota

    for work_id, selection_quota
    in work_selection_quota.items()
)


all_structural_coverage_preserved_gate = (
    len(
        selected_sample_missing_requirements
    )
    == 0
)


every_requirement_has_positive_selected_count_gate = all(
    selected_sample_requirement_counts[
        requirement
    ]
    >= 1

    for requirement
    in sampling_coverage_requirements
)


deterministic_selected_set_recomputes_gate = (
    recomputed_selected_ids
    ==
    selected_exercise_population_unit_ids
)


deterministic_excluded_set_recomputes_gate = (
    recomputed_excluded_ids
    ==
    excluded_exercise_population_unit_ids
)


sample_ids_unique_gate = (
    len(
        {
            record[
                "population_unit_id"
            ]
            for record
            in selected_exercise_unit_sample_records
        }
    )
    == 38
)


all_sample_members_final_membership_false_gate = all(
    record[
        "final_golden_membership"
    ]
    is False

    for record
    in selected_exercise_unit_sample_records
)


legacy_supersession_not_started_gate = (
    EXERCISE_UNIT_REBUILT_SAMPLE_V1_DRAFT[
        "legacy_case_supersession_status"
    ]
    ==
    "NOT_STARTED"
)


target_repair_not_started_gate = (
    EXERCISE_UNIT_REBUILT_SAMPLE_V1_DRAFT[
        "target_repair_status"
    ]
    ==
    "NOT_STARTED"
)


pilot_replacement_not_started_gate = (
    EXERCISE_UNIT_REBUILT_SAMPLE_V1_DRAFT[
        "pilot_replacement_status"
    ]
    ==
    "NOT_STARTED"
)


question_generation_not_started_gate = (
    EXERCISE_UNIT_REBUILT_SAMPLE_V1_DRAFT[
        "question_generation_status"
    ]
    ==
    "NOT_STARTED"
)


expected_answer_not_started_gate = (
    EXERCISE_UNIT_REBUILT_SAMPLE_V1_DRAFT[
        "expected_answer_generation_status"
    ]
    ==
    "NOT_STARTED"
)


no_artifact_write_gate = (
    EXERCISE_UNIT_REBUILT_SAMPLE_V1_DRAFT[
        "artifact_write_status"
    ]
    ==
    "NONE"
)


cell_45_gates = {
    "largest_remainder_derives_exact_5_plus_2_exclusions":
        largest_remainder_allocation_gate,

    "resulting_work_sample_quota_is_exact_26_plus_12":
        selection_work_quota_gate,

    "exact_7_exercise_units_excluded":
        exact_7_excluded_gate,

    "exact_38_exercise_units_selected":
        exact_38_selected_gate,

    "selected_and_excluded_sets_exactly_partition_population":
        selected_excluded_partition_gate,

    "all_work_exclusion_quotas_fully_satisfied":
        all_work_exclusion_quotas_satisfied_gate,

    "excluded_work_counts_match_derived_quota":
        excluded_work_counts_match_quota_gate,

    "selected_work_counts_match_derived_quota":
        selected_work_counts_match_quota_gate,

    "all_cell44_structural_coverage_requirements_preserved":
        all_structural_coverage_preserved_gate,

    "every_structural_requirement_has_positive_selected_representation":
        every_requirement_has_positive_selected_count_gate,

    "selected_set_independently_deterministically_recomputes":
        deterministic_selected_set_recomputes_gate,

    "excluded_set_independently_deterministically_recomputes":
        deterministic_excluded_set_recomputes_gate,

    "all_38_selected_population_ids_unique":
        sample_ids_unique_gate,

    "final_golden_membership_not_yet_applied":
        all_sample_members_final_membership_false_gate,

    "legacy_case_supersession_not_started":
        legacy_supersession_not_started_gate,

    "target_repair_not_started":
        target_repair_not_started_gate,

    "pilot_replacement_not_started":
        pilot_replacement_not_started_gate,

    "question_generation_not_started":
        question_generation_not_started_gate,

    "expected_answer_generation_not_started":
        expected_answer_not_started_gate,

    "no_artifact_write":
        no_artifact_write_gate,
}


cell_45_pass = all(
    cell_45_gates.values()
)


# ============================================================
# 18) Output
# ============================================================

print(
    "=== DETERMINISTIC REBUILT EXERCISE SAMPLING ==="
)

print(
    "Population:",
    EXERCISE_POPULATION_SIZE
)

print(
    "Selected:",
    len(
        selected_exercise_unit_sample_records
    )
)

print(
    "Excluded:",
    len(
        excluded_exercise_unit_rows
    )
)

print(
    "Sample snapshot SHA256:",
    exercise_sample_snapshot_sha256
)


print(
    "\n=== LARGEST-REMAINDER WORK ALLOCATION ==="
)

for row in (
    work_exclusion_allocation_rows
):

    print(
        "- work:",
        row[
            "work_id"
        ],
        "| population:",
        row[
            "population_count"
        ],
        "| exact exclusions:",
        row[
            "exact_exclusion_count"
        ],
        "| remainder:",
        row[
            "fractional_remainder"
        ],
        "| final exclusion quota:",
        row[
            "final_exclusion_quota"
        ],
        "| final selection quota:",
        work_selection_quota[
            row[
                "work_id"
            ]
        ],
    )


print(
    "\n=== SELECTED WORK COUNTS ==="
)

print(
    dict(
        selected_sample_work_counts
    )
)


print(
    "\n=== EXCLUDED WORK COUNTS ==="
)

print(
    dict(
        excluded_sample_work_counts
    )
)


print(
    "\n=== SELECTED STRUCTURAL COVERAGE ==="
)

print(
    "Marker signatures:"
)

for signature, count in sorted(
    selected_sample_marker_signature_counts.items(),
    key=lambda item:
        str(
            item[
                0
            ]
        ),
):

    print(
        "-",
        signature,
        ":",
        count,
    )


print(
    "Table presence:",
    dict(
        selected_sample_table_presence_counts
    )
)


print(
    "Boundary source:",
    dict(
        selected_sample_boundary_source_counts
    )
)


print(
    "Missing Cell44 requirements:",
    len(
        selected_sample_missing_requirements
    )
)


print(
    "\n=== 7 EXCLUDED UNITS ==="
)

for index, row in enumerate(
    excluded_exercise_unit_rows,
    start=1,
):

    print(
        "\nEXCLUDED",
        index,
        "/",
        len(
            excluded_exercise_unit_rows
        )
    )

    print(
        "Population unit:",
        row[
            "population_unit_id"
        ]
    )

    print(
        "Work:",
        row[
            "work_id"
        ]
    )

    print(
        "Heading:",
        row[
            "heading_path"
        ]
    )

    print(
        "Markers:",
        row[
            "terminal_markers"
        ]
    )

    print(
        "Table:",
        row[
            "has_table_record"
        ]
    )

    print(
        "Boundary:",
        row[
            "boundary_source"
        ]
    )

    print(
        "Core tokens:",
        row[
            "exercise_core_token_count"
        ]
    )

    print(
        "Exclusion hash:",
        row[
            "exclusion_hash"
        ]
    )

    print(
        "Reason:",
        row[
            "selection_reason"
        ]
    )


print(
    "\n=== COVERAGE-PROTECTED SKIPS ==="
)

print(
    "Units skipped because exclusion would break coverage:",
    len(
        coverage_protected_skip_rows
    )
)


for row in (
    coverage_protected_skip_rows
):

    print(
        "-",
        row[
            "population_unit_id"
        ],
        "| heading:",
        row[
            "heading_path"
        ],
        "| blocking:",
        row[
            "blocking_requirements"
        ],
    )


print(
    "\n=== LEGACY REPRESENTATION AFTER REBUILD ==="
)

print(
    "Selected units represented in legacy sample:",
    len(
        selected_sample_legacy_represented_ids
    )
)

print(
    "Selected newly discovered units:",
    len(
        selected_sample_new_discovery_ids
    )
)

print(
    "Excluded legacy-represented units:",
    len(
        excluded_legacy_represented_ids
    )
)

print(
    "Excluded newly discovered units:",
    len(
        excluded_new_discovery_ids
    )
)


print(
    "\n=== CODE CELL 45 GATES ==="
)

for name, passed in (
    cell_45_gates.items()
):

    print(
        name,
        ":",
        "PASS"
        if passed
        else "FAIL"
    )


print(
    "\nCODE CELL 45 GATE:",
    "PASS"
    if cell_45_pass
    else "FAIL"
)


assert cell_45_pass, (
    "Deterministic rebuilt Exercise sampling failed. "
    "Do not supersede legacy exercise cases or rebuild Target "
    "Evidence until the exact 38-of-45 sample is deterministic, "
    "work-proportional and independently coverage-valid."
)
```

```python
# Code Cell 46

# ============================================================
# Legacy Exercise Golden Case Downstream Reference Discovery
#
# READ-ONLY DEPENDENCY DISCOVERY.
#
# Scan all pre-existing Python container globals for exact
# references to the 38 legacy exercise Golden Case IDs.
#
# NO mutation.
# NO supersession.
# NO new Case IDs.
# NO Target repair.
# NO Pilot replacement.
# NO question generation.
# NO artifact writes.
# ============================================================


# ============================================================
# 1) Snapshot globals BEFORE creating Cell 46 audit objects
#
# This prevents Cell 46's own newly created variables from
# polluting the reference inventory.
# ============================================================

cell46_preexisting_globals = {
    name:
        value

    for name, value
    in globals().items()

    if not (
        name.startswith(
            "__"
        )
    )
}


# ============================================================
# 2) Exact legacy exercise Golden Case ID universe
# ============================================================

cell46_legacy_exercise_case_ids = {
    row[
        "legacy_golden_case_id"
    ]

    for row
    in legacy_38_to_population_mapping
}


assert (
    len(
        cell46_legacy_exercise_case_ids
    )
    == 38
)


# ============================================================
# 3) Known categories
#
# Classification is about audit provenance only.
# It does NOT imply mutation policy.
# ============================================================

CELL46_LEGACY_SOURCE_VARIABLES = {
    "exercise_selected_cases",
    "exercise_selected_structure_rows",
}


CELL46_REPAIR_TRACEABILITY_VARIABLES = {
    "legacy_exercise_cases_by_group",
    "legacy_to_proposed_exercise_case_mapping",
    "proposed_exercise_unit_cases",
    "proposed_exercise_unit_cases_by_id",
    "proposed_exercise_unit_cases_by_group",
    "proposed_units_with_zero_legacy_inside_core",
    "proposed_units_with_legacy_outside_core",
    "legacy_case_mapping_counts",
    "reconstructed_18_to_population_mapping",
    "legacy_38_to_population_mapping",
    "legacy_represented_population_ids",
    "proposed_case_id_to_population_id",
    "selected_sample_legacy_represented_ids",
    "excluded_legacy_represented_ids",
}


# ============================================================
# 4) Safe recursive exact-ID scanner
#
# Descends only through standard Python containers.
# Objects such as tokenizers, modules, functions, Paths, etc.
# are not traversed.
#
# Cycles are protected through object identity.
# ============================================================

CELL46_SCANNABLE_CONTAINER_TYPES = (
    dict,
    list,
    tuple,
    set,
    frozenset,
)


def cell46_safe_path_key(
    key,
):

    key_repr = repr(
        key
    )

    if (
        len(
            key_repr
        )
        > 180
    ):

        key_repr = (
            key_repr[
                :177
            ]
            +
            "..."
        )

    return key_repr


def cell46_find_legacy_id_references(
    obj,
    target_ids,
    path="$",
    seen=None,
):

    if (
        seen
        is None
    ):

        seen = set()


    # --------------------------------------------------------
    # Exact string match
    # --------------------------------------------------------

    if isinstance(
        obj,
        str,
    ):

        if (
            obj
            in
            target_ids
        ):

            return [
                {
                    "path":
                        path,

                    "legacy_golden_case_id":
                        obj,
                }
            ]

        return []


    # --------------------------------------------------------
    # Non-container objects are out of scope
    # --------------------------------------------------------

    if not isinstance(
        obj,
        CELL46_SCANNABLE_CONTAINER_TYPES,
    ):

        return []


    object_identity = id(
        obj
    )


    if (
        object_identity
        in
        seen
    ):

        return []


    seen.add(
        object_identity
    )


    hits = []


    # --------------------------------------------------------
    # Dict
    # --------------------------------------------------------

    if isinstance(
        obj,
        dict,
    ):

        for key, value in (
            obj.items()
        ):

            # Exact legacy ID used as dictionary key
            if (
                isinstance(
                    key,
                    str,
                )

                and

                key
                in
                target_ids
            ):

                hits.append(
                    {
                        "path":
                            (
                                path
                                +
                                "[DICT_KEY:"
                                +
                                repr(
                                    key
                                )
                                +
                                "]"
                            ),

                        "legacy_golden_case_id":
                            key,
                    }
                )


            value_path = (
                path
                +
                "["
                +
                cell46_safe_path_key(
                    key
                )
                +
                "]"
            )


            hits.extend(
                cell46_find_legacy_id_references(
                    value,
                    target_ids,
                    path=value_path,
                    seen=seen,
                )
            )


    # --------------------------------------------------------
    # List / Tuple
    # --------------------------------------------------------

    elif isinstance(
        obj,
        (
            list,
            tuple,
        ),
    ):

        for index, value in enumerate(
            obj
        ):

            hits.extend(
                cell46_find_legacy_id_references(
                    value,
                    target_ids,
                    path=(
                        path
                        +
                        "["
                        +
                        str(
                            index
                        )
                        +
                        "]"
                    ),
                    seen=seen,
                )
            )


    # --------------------------------------------------------
    # Set / Frozenset
    #
    # Path uses deterministic value representation.
    # --------------------------------------------------------

    elif isinstance(
        obj,
        (
            set,
            frozenset,
        ),
    ):

        for value in sorted(
            obj,
            key=lambda item:
                repr(
                    item
                ),
        ):

            hits.extend(
                cell46_find_legacy_id_references(
                    value,
                    target_ids,
                    path=(
                        path
                        +
                        "[SET:"
                        +
                        cell46_safe_path_key(
                            value
                        )
                        +
                        "]"
                    ),
                    seen=seen,
                )
            )


    return hits


# ============================================================
# 5) Scan all pre-existing top-level globals
# ============================================================

cell46_reference_inventory = []


for variable_name, variable_value in sorted(
    cell46_preexisting_globals.items(),
    key=lambda item:
        item[
            0
        ],
):

    if not isinstance(
        variable_value,
        (
            str,
            dict,
            list,
            tuple,
            set,
            frozenset,
        ),
    ):

        continue


    hits = (
        cell46_find_legacy_id_references(
            variable_value,
            cell46_legacy_exercise_case_ids,
            path="$",
            seen=set(),
        )
    )


    if not hits:
        continue


    distinct_ids = sorted(
        {
            hit[
                "legacy_golden_case_id"
            ]
            for hit
            in hits
        }
    )


    # --------------------------------------------------------
    # Audit classification
    # --------------------------------------------------------

    if (
        variable_name
        in
        CELL46_LEGACY_SOURCE_VARIABLES
    ):

        reference_class = (
            "LEGACY_SOURCE_SAMPLE"
        )


    elif (
        variable_name
        in
        CELL46_REPAIR_TRACEABILITY_VARIABLES
    ):

        reference_class = (
            "REPAIR_OR_TRACEABILITY_STRUCTURE"
        )


    else:

        reference_class = (
            "POTENTIAL_DOWNSTREAM_DEPENDENCY"
        )


    cell46_reference_inventory.append(
        {
            "variable_name":
                variable_name,

            "reference_class":
                reference_class,

            "top_level_type":
                type(
                    variable_value
                ).__name__,

            "reference_count":
                len(
                    hits
                ),

            "distinct_legacy_case_id_count":
                len(
                    distinct_ids
                ),

            "distinct_legacy_case_ids":
                distinct_ids,

            "hits":
                hits,
        }
    )


# ============================================================
# 6) Classify inventory
# ============================================================

cell46_source_reference_rows = [
    row
    for row
    in cell46_reference_inventory
    if (
        row[
            "reference_class"
        ]
        ==
        "LEGACY_SOURCE_SAMPLE"
    )
]


cell46_repair_reference_rows = [
    row
    for row
    in cell46_reference_inventory
    if (
        row[
            "reference_class"
        ]
        ==
        "REPAIR_OR_TRACEABILITY_STRUCTURE"
    )
]


cell46_potential_downstream_reference_rows = [
    row
    for row
    in cell46_reference_inventory
    if (
        row[
            "reference_class"
        ]
        ==
        "POTENTIAL_DOWNSTREAM_DEPENDENCY"
    )
]


# ============================================================
# 7) Distinct IDs represented by category
# ============================================================

def cell46_distinct_ids_for_rows(
    rows,
):

    return {
        legacy_case_id

        for row in rows

        for legacy_case_id
        in row[
            "distinct_legacy_case_ids"
        ]
    }


cell46_source_distinct_ids = (
    cell46_distinct_ids_for_rows(
        cell46_source_reference_rows
    )
)


cell46_repair_distinct_ids = (
    cell46_distinct_ids_for_rows(
        cell46_repair_reference_rows
    )
)


cell46_potential_downstream_distinct_ids = (
    cell46_distinct_ids_for_rows(
        cell46_potential_downstream_reference_rows
    )
)


# ============================================================
# 8) Field/path signal inventory
#
# These are descriptive signals only.
# They help identify likely Task / Target / Pilot dependencies
# without assuming a specific registry variable name.
# ============================================================

CELL46_PATH_SIGNAL_TERMS = {
    "TASK":
        (
            "task",
            "Task",
            "TASK",
        ),

    "TARGET":
        (
            "target",
            "Target",
            "TARGET",
        ),

    "PILOT":
        (
            "pilot",
            "Pilot",
            "PILOT",
        ),

    "QUESTION":
        (
            "question",
            "Question",
            "QUESTION",
        ),

    "ANSWER":
        (
            "answer",
            "Answer",
            "ANSWER",
        ),

    "ADJUDICATION":
        (
            "adjud",
            "Adjud",
            "ADJUD",
        ),

    "PLAN":
        (
            "plan",
            "Plan",
            "PLAN",
        ),

    "CASE":
        (
            "case",
            "Case",
            "CASE",
        ),
}


cell46_downstream_path_signal_counts = Counter()

cell46_downstream_path_signal_examples = defaultdict(
    list
)


for row in (
    cell46_potential_downstream_reference_rows
):

    for hit in (
        row[
            "hits"
        ]
    ):

        combined_text = (
            row[
                "variable_name"
            ]
            +
            " "
            +
            hit[
                "path"
            ]
        )


        for signal_name, terms in (
            CELL46_PATH_SIGNAL_TERMS.items()
        ):

            if any(
                term
                in
                combined_text

                for term
                in terms
            ):

                cell46_downstream_path_signal_counts[
                    signal_name
                ] += 1


                if (
                    len(
                        cell46_downstream_path_signal_examples[
                            signal_name
                        ]
                    )
                    <
                    10
                ):

                    cell46_downstream_path_signal_examples[
                        signal_name
                    ].append(
                        {
                            "variable_name":
                                row[
                                    "variable_name"
                                ],

                            "path":
                                hit[
                                    "path"
                                ],

                            "legacy_golden_case_id":
                                hit[
                                    "legacy_golden_case_id"
                                ],
                        }
                    )


# ============================================================
# 9) Per-legacy-case dependency counts
# ============================================================

cell46_downstream_reference_count_by_legacy_case = Counter()


for row in (
    cell46_potential_downstream_reference_rows
):

    for hit in (
        row[
            "hits"
        ]
    ):

        cell46_downstream_reference_count_by_legacy_case[
            hit[
                "legacy_golden_case_id"
            ]
        ] += 1


cell46_legacy_cases_with_downstream_refs = {
    legacy_case_id

    for legacy_case_id, count
    in cell46_downstream_reference_count_by_legacy_case.items()

    if (
        count
        > 0
    )
}


cell46_legacy_cases_without_downstream_refs = sorted(
    cell46_legacy_exercise_case_ids
    -
    cell46_legacy_cases_with_downstream_refs
)


# ============================================================
# 10) Detect references to the failed pilot legacy case
#
# This case was identified in Cell 31.
# We do NOT assume a registry location here.
# ============================================================

CELL46_FAILED_EXERCISE_PILOT_LEGACY_CASE_ID = (
    "golden-case:49987293a62a89dcd13e121c0a93de1f"
)


cell46_failed_pilot_reference_rows = []


for row in (
    cell46_potential_downstream_reference_rows
):

    matching_hits = [
        hit
        for hit
        in row[
            "hits"
        ]
        if (
            hit[
                "legacy_golden_case_id"
            ]
            ==
            CELL46_FAILED_EXERCISE_PILOT_LEGACY_CASE_ID
        )
    ]


    if (
        matching_hits
    ):

        cell46_failed_pilot_reference_rows.append(
            {
                "variable_name":
                    row[
                        "variable_name"
                    ],

                "matching_hits":
                    matching_hits,
            }
        )


# ============================================================
# 11) Mutation guardrail snapshot
# ============================================================

cell46_selected_sample_ids_before = set(
    selected_exercise_population_unit_ids
)


cell46_legacy_mapping_snapshot = [
    dict(
        row
    )
    for row
    in legacy_38_to_population_mapping
]


# ============================================================
# 12) Audit registry
# ============================================================

LEGACY_EXERCISE_DOWNSTREAM_REFERENCE_AUDIT_V1_DRAFT = {
    "version":
        "legacy_exercise_downstream_reference_audit_v1_draft",

    "status":
        "REFERENCE_DISCOVERY_COMPLETE_NO_MUTATION",

    "legacy_exercise_case_id_count":
        len(
            cell46_legacy_exercise_case_ids
        ),

    "top_level_variables_with_any_legacy_reference":
        len(
            cell46_reference_inventory
        ),

    "legacy_source_variable_count":
        len(
            cell46_source_reference_rows
        ),

    "repair_traceability_variable_count":
        len(
            cell46_repair_reference_rows
        ),

    "potential_downstream_dependency_variable_count":
        len(
            cell46_potential_downstream_reference_rows
        ),

    "legacy_cases_with_potential_downstream_reference_count":
        len(
            cell46_legacy_cases_with_downstream_refs
        ),

    "legacy_cases_without_potential_downstream_reference_count":
        len(
            cell46_legacy_cases_without_downstream_refs
        ),

    "failed_pilot_reference_variable_count":
        len(
            cell46_failed_pilot_reference_rows
        ),

    "supersession_status":
        "NOT_STARTED",

    "new_case_materialization_status":
        "NOT_STARTED",

    "target_repair_status":
        "NOT_STARTED",

    "pilot_replacement_status":
        "NOT_STARTED",

    "question_generation_status":
        "NOT_STARTED",

    "expected_answer_generation_status":
        "NOT_STARTED",

    "artifact_write_status":
        "NONE",
}


# ============================================================
# 13) Validation gates
# ============================================================

exact_38_legacy_ids_gate = (
    len(
        cell46_legacy_exercise_case_ids
    )
    == 38
)


all_legacy_ids_present_in_source_or_traceability_gate = (
    (
        cell46_source_distinct_ids
        |
        cell46_repair_distinct_ids
    )
    ==
    cell46_legacy_exercise_case_ids
)


reference_inventory_classes_valid_gate = all(
    row[
        "reference_class"
    ]
    in {
        "LEGACY_SOURCE_SAMPLE",
        "REPAIR_OR_TRACEABILITY_STRUCTURE",
        "POTENTIAL_DOWNSTREAM_DEPENDENCY",
    }

    for row
    in cell46_reference_inventory
)


all_reference_hits_are_exact_legacy_ids_gate = all(
    hit[
        "legacy_golden_case_id"
    ]
    in
    cell46_legacy_exercise_case_ids

    for row
    in cell46_reference_inventory

    for hit
    in row[
        "hits"
    ]
)


selected_sample_unchanged_gate = (
    set(
        selected_exercise_population_unit_ids
    )
    ==
    cell46_selected_sample_ids_before
)


legacy_mapping_unchanged_gate = (
    legacy_38_to_population_mapping
    ==
    cell46_legacy_mapping_snapshot
)


supersession_not_started_gate = (
    LEGACY_EXERCISE_DOWNSTREAM_REFERENCE_AUDIT_V1_DRAFT[
        "supersession_status"
    ]
    ==
    "NOT_STARTED"
)


new_case_materialization_not_started_gate = (
    LEGACY_EXERCISE_DOWNSTREAM_REFERENCE_AUDIT_V1_DRAFT[
        "new_case_materialization_status"
    ]
    ==
    "NOT_STARTED"
)


target_repair_not_started_gate = (
    LEGACY_EXERCISE_DOWNSTREAM_REFERENCE_AUDIT_V1_DRAFT[
        "target_repair_status"
    ]
    ==
    "NOT_STARTED"
)


pilot_replacement_not_started_gate = (
    LEGACY_EXERCISE_DOWNSTREAM_REFERENCE_AUDIT_V1_DRAFT[
        "pilot_replacement_status"
    ]
    ==
    "NOT_STARTED"
)


question_generation_not_started_gate = (
    LEGACY_EXERCISE_DOWNSTREAM_REFERENCE_AUDIT_V1_DRAFT[
        "question_generation_status"
    ]
    ==
    "NOT_STARTED"
)


expected_answer_not_started_gate = (
    LEGACY_EXERCISE_DOWNSTREAM_REFERENCE_AUDIT_V1_DRAFT[
        "expected_answer_generation_status"
    ]
    ==
    "NOT_STARTED"
)


no_artifact_write_gate = (
    LEGACY_EXERCISE_DOWNSTREAM_REFERENCE_AUDIT_V1_DRAFT[
        "artifact_write_status"
    ]
    ==
    "NONE"
)


cell_46_gates = {
    "exact_38_legacy_exercise_case_ids_audited":
        exact_38_legacy_ids_gate,

    "all_38_legacy_ids_preserved_in_source_or_traceability_structures":
        all_legacy_ids_present_in_source_or_traceability_gate,

    "all_reference_inventory_rows_have_valid_classification":
        reference_inventory_classes_valid_gate,

    "all_discovered_hits_are_exact_legacy_exercise_case_ids":
        all_reference_hits_are_exact_legacy_ids_gate,

    "rebuilt_38_unit_sample_unchanged":
        selected_sample_unchanged_gate,

    "legacy_traceability_mapping_unchanged":
        legacy_mapping_unchanged_gate,

    "legacy_supersession_not_started":
        supersession_not_started_gate,

    "new_exercise_case_materialization_not_started":
        new_case_materialization_not_started_gate,

    "target_repair_not_started":
        target_repair_not_started_gate,

    "pilot_replacement_not_started":
        pilot_replacement_not_started_gate,

    "question_generation_not_started":
        question_generation_not_started_gate,

    "expected_answer_generation_not_started":
        expected_answer_not_started_gate,

    "no_artifact_write":
        no_artifact_write_gate,
}


cell_46_pass = all(
    cell_46_gates.values()
)


# ============================================================
# 14) Output
# ============================================================

print(
    "=== LEGACY EXERCISE GOLDEN CASE DOWNSTREAM REFERENCE DISCOVERY ==="
)

print(
    "Legacy Exercise Golden Case IDs:",
    len(
        cell46_legacy_exercise_case_ids
    )
)

print(
    "Top-level variables with legacy references:",
    len(
        cell46_reference_inventory
    )
)


print(
    "\n=== REFERENCE CLASS COUNTS ==="
)

print(
    "Legacy source variables:",
    len(
        cell46_source_reference_rows
    )
)

print(
    "Repair / traceability variables:",
    len(
        cell46_repair_reference_rows
    )
)

print(
    "Potential downstream dependency variables:",
    len(
        cell46_potential_downstream_reference_rows
    )
)


print(
    "\n=== LEGACY SOURCE REFERENCES ==="
)

for row in (
    cell46_source_reference_rows
):

    print(
        "-",
        row[
            "variable_name"
        ],
        "| type:",
        row[
            "top_level_type"
        ],
        "| refs:",
        row[
            "reference_count"
        ],
        "| distinct IDs:",
        row[
            "distinct_legacy_case_id_count"
        ],
    )


print(
    "\n=== REPAIR / TRACEABILITY REFERENCES ==="
)

for row in (
    cell46_repair_reference_rows
):

    print(
        "-",
        row[
            "variable_name"
        ],
        "| refs:",
        row[
            "reference_count"
        ],
        "| distinct IDs:",
        row[
            "distinct_legacy_case_id_count"
        ],
    )


print(
    "\n=== POTENTIAL DOWNSTREAM DEPENDENCY VARIABLES ==="
)

if (
    cell46_potential_downstream_reference_rows
):

    for row in (
        cell46_potential_downstream_reference_rows
    ):

        print(
            "\nVARIABLE:",
            row[
                "variable_name"
            ]
        )

        print(
            "Type:",
            row[
                "top_level_type"
            ]
        )

        print(
            "References:",
            row[
                "reference_count"
            ]
        )

        print(
            "Distinct legacy IDs:",
            row[
                "distinct_legacy_case_id_count"
            ]
        )


        print(
            "Example paths:"
        )

        for hit in (
            row[
                "hits"
            ][
                :20
            ]
        ):

            print(
                "  -",
                hit[
                    "legacy_golden_case_id"
                ],
                "|",
                hit[
                    "path"
                ],
            )

else:

    print(
        "NONE"
    )


print(
    "\n=== POTENTIAL DOWNSTREAM REFERENCE COVERAGE ==="
)

print(
    "Legacy IDs with at least one potential downstream reference:",
    len(
        cell46_legacy_cases_with_downstream_refs
    )
)

print(
    "Legacy IDs with no potential downstream reference:",
    len(
        cell46_legacy_cases_without_downstream_refs
    )
)


if (
    cell46_legacy_cases_without_downstream_refs
):

    for legacy_case_id in (
        cell46_legacy_cases_without_downstream_refs
    ):

        print(
            "- no downstream ref:",
            legacy_case_id
        )


print(
    "\n=== PATH SIGNAL COUNTS ==="
)

if (
    cell46_downstream_path_signal_counts
):

    for signal_name, count in sorted(
        cell46_downstream_path_signal_counts.items()
    ):

        print(
            "-",
            signal_name,
            ":",
            count,
        )

else:

    print(
        "NONE"
    )


print(
    "\n=== PATH SIGNAL EXAMPLES ==="
)

for signal_name in sorted(
    cell46_downstream_path_signal_examples
):

    print(
        "\n",
        signal_name,
        sep="",
    )

    for example in (
        cell46_downstream_path_signal_examples[
            signal_name
        ]
    ):

        print(
            "-",
            example[
                "variable_name"
            ],
            "|",
            example[
                "legacy_golden_case_id"
            ],
            "|",
            example[
                "path"
            ],
        )


print(
    "\n=== FAILED EXERCISE PILOT LEGACY CASE REFERENCES ==="
)

print(
    "Failed pilot legacy case:",
    CELL46_FAILED_EXERCISE_PILOT_LEGACY_CASE_ID
)

print(
    "Potential downstream variables referencing it:",
    len(
        cell46_failed_pilot_reference_rows
    )
)


for row in (
    cell46_failed_pilot_reference_rows
):

    print(
        "-",
        row[
            "variable_name"
        ]
    )

    for hit in (
        row[
            "matching_hits"
        ][
            :20
        ]
    ):

        print(
            "  ",
            hit[
                "path"
            ]
        )


print(
    "\n=== CODE CELL 46 GATES ==="
)

for name, passed in (
    cell_46_gates.items()
):

    print(
        name,
        ":",
        "PASS"
        if passed
        else "FAIL"
    )


print(
    "\nCODE CELL 46 GATE:",
    "PASS"
    if cell_46_pass
    else "FAIL"
)


assert cell_46_pass, (
    "Legacy downstream dependency discovery failed. "
    "Do not supersede legacy exercise cases until every "
    "in-memory reference to the 38 legacy Golden Case IDs "
    "has been inventoried without mutation."
)
```

<h3>Canonical Downstream Dependency and Alias Audit</h3>

<p>
Cell 46 نشان داد 45 Variable در حافظه به 38 Legacy Exercise Golden Case اشاره می‌کنند.
اما تعداد Variable الزاماً برابر تعداد Dependency Object واقعی نیست.
</p>

<p>
در Notebookهای تحلیلی ممکن است چند Variable به یک List یا Dict مشترک اشاره کنند،
یا Variableهای موقتی Loop فقط آخرین Record یک Registry را نگه داشته باشند.
اگر این تفاوت در Repair لحاظ نشود، ممکن است یک Object چند بار Repair شود
یا Snapshotهای تاریخی با Registryهای فعال اشتباه گرفته شوند.
</p>

<p>
این Cell سه کار انجام می‌دهد:
</p>

<ul>
<li>
Variableهای Cell 46 را بر اساس Python Object Identity یعنی <code>id()</code>
deduplicate می‌کند؛
</li>

<li>
Schema واقعی Registryهای اصلی Exercise را بدون فرض‌کردن Fieldهای ناموجود استخراج می‌کند؛
</li>

<li>
زنجیره‌ی فعلی
<strong>Golden Case → Task Draft → Target Evidence → Pilot</strong>
را از روی IDها و Fieldهای واقعی Notebook Audit می‌کند.
</li>
</ul>

<p>
Registryهای تاریخی مربوط به Boundary Discovery، Evidence Scope Audit و Repair Traceability
در این مرحله فقط به‌عنوان Historical/Audit State شناخته می‌شوند.
آن‌ها نباید هنگام Materialization جدید به‌صورت کور Rewrite شوند.
</p>

<p>
هیچ Case، Task، Target یا Pilot در این Cell تغییر نمی‌کند.
هدف فقط ساخت یک Repair Scope دقیق و مستقل از Aliasهاست.
</p>

</div>

```python
# Code Cell 47

# ============================================================
# Canonical Downstream Dependency & Alias Audit
#
# INPUT:
#   Cell 46 reference inventory
#
# PURPOSE:
#   1) deduplicate top-level variables by Python object identity
#   2) distinguish durable registries from aliases / snapshots
#   3) inspect actual schemas
#   4) reconstruct the current
#        Golden Case -> Task Draft -> Target Evidence -> Pilot
#      dependency chain for the 38 legacy exercise cases
#
# READ-ONLY.
#
# NO supersession.
# NO new case materialization.
# NO Task rebuild.
# NO Target rebuild.
# NO Pilot replacement.
# NO question generation.
# NO artifact writes.
# ============================================================


# ============================================================
# 1) Snapshot the 45 Cell-46 potential dependency variables
# ============================================================

cell47_dependency_variable_names = [
    row[
        "variable_name"
    ]
    for row
    in cell46_potential_downstream_reference_rows
]


cell47_dependency_variables = {
    variable_name:
        cell46_preexisting_globals[
            variable_name
        ]

    for variable_name
    in cell47_dependency_variable_names
}


assert (
    len(
        cell47_dependency_variables
    )
    ==
    len(
        cell46_potential_downstream_reference_rows
    )
)


# ============================================================
# 2) Deduplicate by Python object identity
#
# Different top-level variable names can point to the same
# underlying list/dict/string object.
# ============================================================

cell47_object_groups = defaultdict(
    list
)


for variable_name, obj in (
    cell47_dependency_variables.items()
):

    cell47_object_groups[
        id(
            obj
        )
    ].append(
        variable_name
    )


cell47_unique_object_rows = []


for object_id, variable_names in sorted(
    cell47_object_groups.items(),
    key=lambda item:
        min(
            item[
                1
            ]
        ),
):

    representative_name = sorted(
        variable_names
    )[0]


    representative_obj = (
        cell47_dependency_variables[
            representative_name
        ]
    )


    reference_inventory_rows = [
        row
        for row
        in cell46_potential_downstream_reference_rows
        if (
            row[
                "variable_name"
            ]
            in
            variable_names
        )
    ]


    distinct_legacy_ids = sorted(
        {
            legacy_case_id

            for row
            in reference_inventory_rows

            for legacy_case_id
            in row[
                "distinct_legacy_case_ids"
            ]
        }
    )


    cell47_unique_object_rows.append(
        {
            "object_id":
                object_id,

            "representative_variable":
                representative_name,

            "variable_names":
                sorted(
                    variable_names
                ),

            "alias_count":
                len(
                    variable_names
                ),

            "object_type":
                type(
                    representative_obj
                ).__name__,

            "distinct_legacy_case_id_count":
                len(
                    distinct_legacy_ids
                ),

            "distinct_legacy_case_ids":
                distinct_legacy_ids,
        }
    )


cell47_alias_groups = [
    row
    for row
    in cell47_unique_object_rows
    if (
        row[
            "alias_count"
        ]
        > 1
    )
]


# ============================================================
# 3) Define only the known durable registry candidates that
# were actually discovered by Cell 46.
#
# Inclusion here does NOT mean "rewrite".
# It only means "inspect as a likely durable dependency".
# ============================================================

CELL47_DURABLE_REGISTRY_CANDIDATES = [
    "golden_case_records",
    "golden_case_records_by_id",
    "auto_case_records",
    "case_groundability_precheck",
    "case_task_design",
    "ready_case_plans",
    "task_draft_plan_rows",
    "task_draft_plans",
    "target_evidence_records",
    "target_evidence_by_case_id",
    "target_evidence_by_task_draft_id",
    "targets_by_stratum",
    "pilot_selection_rows",
    "pilot_full_inspection_rows",
    "pilot_semantic_review_rows",
    "question_drafting_pilot_packets",
    "question_drafting_pilot_by_id",
]


cell47_present_durable_registries = [
    variable_name
    for variable_name
    in CELL47_DURABLE_REGISTRY_CANDIDATES
    if (
        variable_name
        in
        cell47_dependency_variables
    )
]


# ============================================================
# 4) Historical / diagnostic structures that should be
# preserved as audit evidence rather than silently rewritten.
#
# Again: classification only; no mutation.
# ============================================================

CELL47_HISTORICAL_AUDIT_CANDIDATES = [
    "exercise_evidence_scope_audit_rows",
    "exercise_group_boundary_discovery_rows",
    "exercise_method_audit_rows",
    "groups_with_multiple_selected_cases",
    "selected_cases_by_exercise_group",
    "failed_exercise_audit",
    "failed_exercise_scope_audit",
    "failed_exercise_structure_row",
    "failed_exercise_pilot",
    "repair_rows",
]


cell47_present_historical_audit_variables = [
    variable_name
    for variable_name
    in CELL47_HISTORICAL_AUDIT_CANDIDATES
    if (
        variable_name
        in
        cell47_dependency_variables
    )
]


# ============================================================
# 5) Likely ephemeral scalar / last-loop aliases
#
# We classify only by object shape and tiny legacy-ID coverage.
# No deletion is performed.
# ============================================================

cell47_ephemeral_alias_candidates = []


for row in (
    cell47_unique_object_rows
):

    representative_name = (
        row[
            "representative_variable"
        ]
    )


    if (
        representative_name
        in
        cell47_present_durable_registries

        or

        representative_name
        in
        cell47_present_historical_audit_variables
    ):

        continue


    if (
        row[
            "distinct_legacy_case_id_count"
        ]
        <= 4

        and

        row[
            "object_type"
        ]
        in {
            "dict",
            "list",
            "str",
        }
    ):

        cell47_ephemeral_alias_candidates.append(
            row
        )


# ============================================================
# 6) Generic schema inspector
#
# For list-of-dict / dict-of-dict registries, discover actual
# key unions without assuming schema.
# ============================================================

def cell47_schema_summary(
    obj,
):

    summary = {
        "top_level_type":
            type(
                obj
            ).__name__,

        "top_level_size":
            (
                len(
                    obj
                )
                if isinstance(
                    obj,
                    (
                        dict,
                        list,
                        tuple,
                        set,
                        frozenset,
                    ),
                )
                else None
            ),

        "record_count_inspected":
            0,

        "record_key_union":
            [],

        "record_key_intersection":
            [],

        "value_types":
            {},
    }


    records = []


    if isinstance(
        obj,
        list,
    ):

        records = [
            value
            for value
            in obj
            if isinstance(
                value,
                dict,
            )
        ]


    elif isinstance(
        obj,
        dict,
    ):

        records = [
            value
            for value
            in obj.values()
            if isinstance(
                value,
                dict,
            )
        ]


    if not records:

        return summary


    key_sets = [
        set(
            record.keys()
        )
        for record
        in records
    ]


    key_union = set().union(
        *key_sets
    )


    key_intersection = set(
        key_sets[
            0
        ]
    )


    for key_set in (
        key_sets[
            1:
        ]
    ):

        key_intersection &= (
            key_set
        )


    value_type_sets = defaultdict(
        set
    )


    for record in (
        records
    ):

        for key, value in (
            record.items()
        ):

            value_type_sets[
                key
            ].add(
                type(
                    value
                ).__name__
            )


    summary[
        "record_count_inspected"
    ] = len(
        records
    )


    summary[
        "record_key_union"
    ] = sorted(
        key_union
    )


    summary[
        "record_key_intersection"
    ] = sorted(
        key_intersection
    )


    summary[
        "value_types"
    ] = {
        key:
            sorted(
                value_types
            )

        for key, value_types
        in sorted(
            value_type_sets.items()
        )
    }


    return summary


# ============================================================
# 7) Extract only records belonging to the 38 legacy exercise
# cases from a generic registry.
# ============================================================

def cell47_extract_legacy_records(
    obj,
    legacy_case_ids,
):

    extracted = []


    if isinstance(
        obj,
        list,
    ):

        for index, value in enumerate(
            obj
        ):

            if not isinstance(
                value,
                dict,
            ):

                continue


            golden_case_id = (
                value.get(
                    "golden_case_id"
                )
            )


            if (
                golden_case_id
                in
                legacy_case_ids
            ):

                extracted.append(
                    {
                        "location":
                            (
                                "LIST_INDEX",
                                index,
                            ),

                        "record":
                            value,
                    }
                )


    elif isinstance(
        obj,
        dict,
    ):

        for key, value in (
            obj.items()
        ):

            if (
                isinstance(
                    key,
                    str,
                )

                and

                key
                in
                legacy_case_ids
            ):

                extracted.append(
                    {
                        "location":
                            (
                                "DICT_KEY",
                                key,
                            ),

                        "record":
                            value,
                    }
                )

                continue


            if not isinstance(
                value,
                dict,
            ):

                continue


            golden_case_id = (
                value.get(
                    "golden_case_id"
                )
            )


            if (
                golden_case_id
                in
                legacy_case_ids
            ):

                extracted.append(
                    {
                        "location":
                            (
                                "DICT_VALUE",
                                key,
                            ),

                        "record":
                            value,
                    }
                )


    return extracted


# ============================================================
# 8) Durable-registry schema and legacy-subset audit
# ============================================================

cell47_durable_registry_audit_rows = []


for variable_name in (
    cell47_present_durable_registries
):

    obj = (
        cell47_dependency_variables[
            variable_name
        ]
    )


    schema_summary = (
        cell47_schema_summary(
            obj
        )
    )


    legacy_records = (
        cell47_extract_legacy_records(
            obj,
            cell46_legacy_exercise_case_ids,
        )
    )


    distinct_legacy_ids = sorted(
        {
            (
                item[
                    "record"
                ].get(
                    "golden_case_id"
                )
                if isinstance(
                    item[
                        "record"
                    ],
                    dict,
                )
                else None
            )

            for item
            in legacy_records
        }
        -
        {
            None
        }
    )


    cell47_durable_registry_audit_rows.append(
        {
            "variable_name":
                variable_name,

            "object_id":
                id(
                    obj
                ),

            "schema":
                schema_summary,

            "legacy_record_hits":
                len(
                    legacy_records
                ),

            "distinct_legacy_case_id_count":
                len(
                    distinct_legacy_ids
                ),

            "distinct_legacy_case_ids":
                distinct_legacy_ids,

            "legacy_records":
                legacy_records,
        }
    )


# ============================================================
# 9) Actual field inventory for dependency IDs
#
# Discover keys that appear to carry IDs; do not assume their
# exact names in advance.
# ============================================================

CELL47_ID_FIELD_SIGNAL_TERMS = (
    "_id",
    "id",
)


def cell47_id_like_fields_from_schema(
    schema_summary,
):

    fields = []


    for field_name in (
        schema_summary[
            "record_key_union"
        ]
    ):

        lowered = (
            str(
                field_name
            ).lower()
        )


        if (
            lowered
            == "id"

            or

            lowered.endswith(
                "_id"
            )

            or

            "_id_" in lowered
        ):

            fields.append(
                field_name
            )


    return sorted(
        fields
    )


cell47_registry_id_field_inventory = {
    row[
        "variable_name"
    ]:
        cell47_id_like_fields_from_schema(
            row[
                "schema"
            ]
        )

    for row
    in cell47_durable_registry_audit_rows
}


# ============================================================
# 10) Build normalized legacy case -> dependency-field map
#
# For every durable registry record referring to a legacy case,
# retain all actual *_id / id-like fields.
# ============================================================

cell47_case_dependency_field_rows = []


for registry_row in (
    cell47_durable_registry_audit_rows
):

    variable_name = (
        registry_row[
            "variable_name"
        ]
    )


    id_fields = (
        cell47_registry_id_field_inventory[
            variable_name
        ]
    )


    for item in (
        registry_row[
            "legacy_records"
        ]
    ):

        record = (
            item[
                "record"
            ]
        )


        if not isinstance(
            record,
            dict,
        ):

            continue


        golden_case_id = (
            record.get(
                "golden_case_id"
            )
        )


        if (
            golden_case_id
            not in
            cell46_legacy_exercise_case_ids
        ):

            continue


        dependency_ids = {
            field_name:
                record.get(
                    field_name
                )

            for field_name
            in id_fields

            if (
                record.get(
                    field_name
                )
                is not None
            )
        }


        cell47_case_dependency_field_rows.append(
            {
                "registry":
                    variable_name,

                "golden_case_id":
                    golden_case_id,

                "dependency_ids":
                    dependency_ids,

                "location":
                    item[
                        "location"
                    ],
            }
        )


# ============================================================
# 11) Per-case durable dependency coverage
# ============================================================

cell47_durable_registries_by_legacy_case = defaultdict(
    set
)


for row in (
    cell47_case_dependency_field_rows
):

    cell47_durable_registries_by_legacy_case[
        row[
            "golden_case_id"
        ]
    ].add(
        row[
            "registry"
        ]
    )


cell47_durable_dependency_count_by_case = {
    legacy_case_id:
        len(
            cell47_durable_registries_by_legacy_case[
                legacy_case_id
            ]
        )

    for legacy_case_id
    in cell46_legacy_exercise_case_ids
}


# ============================================================
# 12) Detect concrete Task Draft IDs
# ============================================================

cell47_task_draft_ids_by_case = defaultdict(
    set
)


for row in (
    cell47_case_dependency_field_rows
):

    for field_name, value in (
        row[
            "dependency_ids"
        ].items()
    ):

        normalized_field = (
            str(
                field_name
            ).lower()
        )


        if (
            "task"
            in
            normalized_field

            and

            "id"
            in
            normalized_field

            and

            isinstance(
                value,
                str,
            )
        ):

            cell47_task_draft_ids_by_case[
                row[
                    "golden_case_id"
                ]
            ].add(
                value
            )


# ============================================================
# 13) Detect concrete Target Evidence IDs
# ============================================================

cell47_target_ids_by_case = defaultdict(
    set
)


for row in (
    cell47_case_dependency_field_rows
):

    for field_name, value in (
        row[
            "dependency_ids"
        ].items()
    ):

        normalized_field = (
            str(
                field_name
            ).lower()
        )


        if (
            "target"
            in
            normalized_field

            and

            "id"
            in
            normalized_field

            and

            isinstance(
                value,
                str,
            )
        ):

            cell47_target_ids_by_case[
                row[
                    "golden_case_id"
                ]
            ].add(
                value
            )


# ============================================================
# 14) Pilot dependency audit
#
# Use actual discovered Pilot-related variables.
# ============================================================

CELL47_PILOT_REGISTRY_NAMES = [
    "pilot_selection_rows",
    "pilot_full_inspection_rows",
    "pilot_semantic_review_rows",
    "question_drafting_pilot_packets",
    "question_drafting_pilot_by_id",
]


cell47_pilot_dependency_rows = []


for variable_name in (
    CELL47_PILOT_REGISTRY_NAMES
):

    if (
        variable_name
        not in
        cell47_dependency_variables
    ):

        continue


    obj = (
        cell47_dependency_variables[
            variable_name
        ]
    )


    legacy_records = (
        cell47_extract_legacy_records(
            obj,
            cell46_legacy_exercise_case_ids,
        )
    )


    for item in (
        legacy_records
    ):

        record = (
            item[
                "record"
            ]
        )


        if not isinstance(
            record,
            dict,
        ):

            continue


        cell47_pilot_dependency_rows.append(
            {
                "registry":
                    variable_name,

                "golden_case_id":
                    record.get(
                        "golden_case_id"
                    ),

                "location":
                    item[
                        "location"
                    ],

                "record_keys":
                    sorted(
                        record.keys()
                    ),

                "id_like_fields":
                    {
                        key:
                            value

                        for key, value
                        in record.items()

                        if (
                            (
                                str(
                                    key
                                ).lower()
                                ==
                                "id"
                            )

                            or

                            str(
                                key
                            ).lower().endswith(
                                "_id"
                            )

                            or

                            "_id_"
                            in
                            str(
                                key
                            ).lower()
                        )
                    },
            }
        )


cell47_pilot_legacy_case_ids = {
    row[
        "golden_case_id"
    ]

    for row
    in cell47_pilot_dependency_rows

    if (
        row[
            "golden_case_id"
        ]
        in
        cell46_legacy_exercise_case_ids
    )
}


# ============================================================
# 15) Determine repair-scope classification by registry
#
# This is a PLAN classification only.
# ============================================================

CELL47_REPAIR_CLASSIFICATION = {
    # Core membership / case registries
    "golden_case_records":
        "REBUILD_EXERCISE_SLICE",

    "golden_case_records_by_id":
        "REBUILD_INDEX_AFTER_CASE_REBUILD",

    "auto_case_records":
        "REBUILD_EXERCISE_SLICE",

    # Case readiness / task design
    "case_groundability_precheck":
        "REBUILD_FOR_NEW_EXERCISE_CASE_IDS",

    "case_task_design":
        "REBUILD_FOR_NEW_EXERCISE_CASE_IDS",

    "ready_case_plans":
        "REBUILD_EXERCISE_SLICE",

    # Task drafts
    "task_draft_plan_rows":
        "REBUILD_EXERCISE_SLICE",

    "task_draft_plans":
        "REBUILD_INDEX_AFTER_TASK_REBUILD",

    # Target evidence
    "target_evidence_records":
        "REBUILD_EXERCISE_SLICE",

    "target_evidence_by_case_id":
        "REBUILD_INDEX_AFTER_TARGET_REBUILD",

    "target_evidence_by_task_draft_id":
        "REBUILD_INDEX_AFTER_TARGET_REBUILD",

    "targets_by_stratum":
        "REBUILD_EXERCISE_STRATUM_SLICE",

    # Pilot
    "pilot_selection_rows":
        "RESELECT_EXERCISE_PILOT",

    "pilot_full_inspection_rows":
        "REBUILD_EXERCISE_PILOT_DERIVATIVE",

    "pilot_semantic_review_rows":
        "REBUILD_EXERCISE_PILOT_DERIVATIVE",

    "question_drafting_pilot_packets":
        "REBUILD_EXERCISE_PILOT_PACKET",

    "question_drafting_pilot_by_id":
        "REBUILD_PILOT_INDEX",
}


cell47_repair_scope_rows = []


for variable_name in (
    cell47_present_durable_registries
):

    cell47_repair_scope_rows.append(
        {
            "variable_name":
                variable_name,

            "planned_action":
                CELL47_REPAIR_CLASSIFICATION.get(
                    variable_name,
                    "INSPECT_BEFORE_DECISION",
                ),

            "object_id":
                id(
                    cell47_dependency_variables[
                        variable_name
                    ]
                ),
        }
    )


# ============================================================
# 16) Alias-aware repair object groups
#
# If multiple durable variables share an object, repair must
# happen once at object level, then indexes/aliases validated.
# ============================================================

cell47_durable_object_groups = defaultdict(
    list
)


for row in (
    cell47_repair_scope_rows
):

    cell47_durable_object_groups[
        row[
            "object_id"
        ]
    ].append(
        row[
            "variable_name"
        ]
    )


cell47_durable_shared_object_rows = [
    {
        "object_id":
            object_id,

        "variables":
            sorted(
                variable_names
            ),
    }

    for object_id, variable_names
    in cell47_durable_object_groups.items()

    if (
        len(
            variable_names
        )
        > 1
    )
]


# ============================================================
# 17) Current-chain integrity observations
# ============================================================

cell47_cases_with_task_ids = {
    legacy_case_id

    for legacy_case_id, values
    in cell47_task_draft_ids_by_case.items()

    if values
}


cell47_cases_with_target_ids = {
    legacy_case_id

    for legacy_case_id, values
    in cell47_target_ids_by_case.items()

    if values
}


cell47_cases_without_task_ids = sorted(
    cell46_legacy_exercise_case_ids
    -
    cell47_cases_with_task_ids
)


cell47_cases_without_target_ids = sorted(
    cell46_legacy_exercise_case_ids
    -
    cell47_cases_with_target_ids
)


# ============================================================
# 18) Audit registry
# ============================================================

EXERCISE_DOWNSTREAM_CANONICAL_DEPENDENCY_AUDIT_V1_DRAFT = {
    "version":
        "exercise_downstream_canonical_dependency_audit_v1_draft",

    "status":
        "CANONICAL_CHAIN_AND_ALIAS_AUDITED_NO_MUTATION",

    "cell46_potential_dependency_variable_count":
        len(
            cell47_dependency_variable_names
        ),

    "unique_underlying_object_count":
        len(
            cell47_unique_object_rows
        ),

    "alias_group_count":
        len(
            cell47_alias_groups
        ),

    "durable_registry_count":
        len(
            cell47_present_durable_registries
        ),

    "historical_audit_variable_count":
        len(
            cell47_present_historical_audit_variables
        ),

    "ephemeral_alias_candidate_count":
        len(
            cell47_ephemeral_alias_candidates
        ),

    "legacy_cases_with_task_id_count":
        len(
            cell47_cases_with_task_ids
        ),

    "legacy_cases_with_target_id_count":
        len(
            cell47_cases_with_target_ids
        ),

    "pilot_legacy_case_count":
        len(
            cell47_pilot_legacy_case_ids
        ),

    "supersession_status":
        "NOT_STARTED",

    "new_case_materialization_status":
        "NOT_STARTED",

    "task_rebuild_status":
        "NOT_STARTED",

    "target_rebuild_status":
        "NOT_STARTED",

    "pilot_rebuild_status":
        "NOT_STARTED",

    "question_generation_status":
        "NOT_STARTED",

    "artifact_write_status":
        "NONE",
}


# ============================================================
# 19) Mutation guardrails
# ============================================================

cell47_selected_sample_snapshot = set(
    selected_exercise_population_unit_ids
)


cell47_golden_case_count_snapshot = len(
    golden_case_records
)


cell47_target_evidence_count_snapshot = len(
    target_evidence_records
)


# ============================================================
# 20) Validation gates
# ============================================================

cell46_45_variables_reproduced_gate = (
    len(
        cell47_dependency_variable_names
    )
    == 45
)


unique_object_accounting_gate = (
    sum(
        row[
            "alias_count"
        ]
        for row
        in cell47_unique_object_rows
    )
    ==
    45
)


all_durable_registry_candidates_present_gate = (
    set(
        cell47_present_durable_registries
    )
    ==
    set(
        CELL47_DURABLE_REGISTRY_CANDIDATES
    )
)


all_durable_repair_actions_classified_gate = all(
    row[
        "planned_action"
    ]
    !=
    "INSPECT_BEFORE_DECISION"

    for row
    in cell47_repair_scope_rows
)


all_38_cases_have_durable_dependency_presence_gate = all(
    cell47_durable_dependency_count_by_case[
        legacy_case_id
    ]
    > 0

    for legacy_case_id
    in cell46_legacy_exercise_case_ids
)


all_38_cases_have_task_ids_gate = (
    len(
        cell47_cases_without_task_ids
    )
    == 0
)


all_38_cases_have_target_ids_gate = (
    len(
        cell47_cases_without_target_ids
    )
    == 0
)


failed_pilot_present_in_pilot_chain_gate = (
    CELL46_FAILED_EXERCISE_PILOT_LEGACY_CASE_ID
    in
    cell47_pilot_legacy_case_ids
)


historical_audit_variables_preserved_gate = (
    len(
        cell47_present_historical_audit_variables
    )
    > 0
)


rebuilt_sample_unchanged_gate = (
    set(
        selected_exercise_population_unit_ids
    )
    ==
    cell47_selected_sample_snapshot
)


golden_case_records_not_mutated_gate = (
    len(
        golden_case_records
    )
    ==
    cell47_golden_case_count_snapshot
)


target_evidence_records_not_mutated_gate = (
    len(
        target_evidence_records
    )
    ==
    cell47_target_evidence_count_snapshot
)


supersession_not_started_gate = (
    EXERCISE_DOWNSTREAM_CANONICAL_DEPENDENCY_AUDIT_V1_DRAFT[
        "supersession_status"
    ]
    ==
    "NOT_STARTED"
)


new_case_materialization_not_started_gate = (
    EXERCISE_DOWNSTREAM_CANONICAL_DEPENDENCY_AUDIT_V1_DRAFT[
        "new_case_materialization_status"
    ]
    ==
    "NOT_STARTED"
)


task_rebuild_not_started_gate = (
    EXERCISE_DOWNSTREAM_CANONICAL_DEPENDENCY_AUDIT_V1_DRAFT[
        "task_rebuild_status"
    ]
    ==
    "NOT_STARTED"
)


target_rebuild_not_started_gate = (
    EXERCISE_DOWNSTREAM_CANONICAL_DEPENDENCY_AUDIT_V1_DRAFT[
        "target_rebuild_status"
    ]
    ==
    "NOT_STARTED"
)


pilot_rebuild_not_started_gate = (
    EXERCISE_DOWNSTREAM_CANONICAL_DEPENDENCY_AUDIT_V1_DRAFT[
        "pilot_rebuild_status"
    ]
    ==
    "NOT_STARTED"
)


question_generation_not_started_gate = (
    EXERCISE_DOWNSTREAM_CANONICAL_DEPENDENCY_AUDIT_V1_DRAFT[
        "question_generation_status"
    ]
    ==
    "NOT_STARTED"
)


no_artifact_write_gate = (
    EXERCISE_DOWNSTREAM_CANONICAL_DEPENDENCY_AUDIT_V1_DRAFT[
        "artifact_write_status"
    ]
    ==
    "NONE"
)


cell_47_gates = {
    "cell46_exact_45_potential_dependency_variables_reproduced":
        cell46_45_variables_reproduced_gate,

    "all_45_variable_names_accounted_for_by_unique_object_groups":
        unique_object_accounting_gate,

    "all_expected_durable_dependency_registries_present":
        all_durable_registry_candidates_present_gate,

    "all_durable_registries_receive_explicit_repair_classification":
        all_durable_repair_actions_classified_gate,

    "all_38_legacy_cases_present_in_durable_dependency_chain":
        all_38_cases_have_durable_dependency_presence_gate,

    "all_38_legacy_cases_have_current_task_ids":
        all_38_cases_have_task_ids_gate,

    "all_38_legacy_cases_have_current_target_ids":
        all_38_cases_have_target_ids_gate,

    "failed_exercise_pilot_present_in_current_pilot_chain":
        failed_pilot_present_in_pilot_chain_gate,

    "historical_audit_structures_identified_for_preservation":
        historical_audit_variables_preserved_gate,

    "rebuilt_38_unit_sample_unchanged":
        rebuilt_sample_unchanged_gate,

    "golden_case_records_not_mutated":
        golden_case_records_not_mutated_gate,

    "target_evidence_records_not_mutated":
        target_evidence_records_not_mutated_gate,

    "legacy_supersession_not_started":
        supersession_not_started_gate,

    "new_exercise_case_materialization_not_started":
        new_case_materialization_not_started_gate,

    "task_rebuild_not_started":
        task_rebuild_not_started_gate,

    "target_rebuild_not_started":
        target_rebuild_not_started_gate,

    "pilot_rebuild_not_started":
        pilot_rebuild_not_started_gate,

    "question_generation_not_started":
        question_generation_not_started_gate,

    "no_artifact_write":
        no_artifact_write_gate,
}


cell_47_pass = all(
    cell_47_gates.values()
)


# ============================================================
# 21) Output
# ============================================================

print(
    "=== CANONICAL DOWNSTREAM DEPENDENCY & ALIAS AUDIT ==="
)

print(
    "Cell46 potential dependency variables:",
    len(
        cell47_dependency_variable_names
    )
)

print(
    "Unique underlying Python objects:",
    len(
        cell47_unique_object_rows
    )
)

print(
    "Alias groups:",
    len(
        cell47_alias_groups
    )
)


print(
    "\n=== ALIAS GROUPS ==="
)

if (
    cell47_alias_groups
):

    for row in (
        cell47_alias_groups
    ):

        print(
            "- object:",
            row[
                "object_id"
            ],
            "| variables:",
            row[
                "variable_names"
            ],
            "| legacy IDs:",
            row[
                "distinct_legacy_case_id_count"
            ],
        )

else:

    print(
        "NONE"
    )


print(
    "\n=== DURABLE REGISTRY SCHEMA AUDIT ==="
)

for row in (
    cell47_durable_registry_audit_rows
):

    variable_name = (
        row[
            "variable_name"
        ]
    )


    repair_action = next(
        item[
            "planned_action"
        ]
        for item
        in cell47_repair_scope_rows
        if (
            item[
                "variable_name"
            ]
            ==
            variable_name
        )
    )


    print(
        "\nREGISTRY:",
        variable_name
    )

    print(
        "Object ID:",
        row[
            "object_id"
        ]
    )

    print(
        "Top-level type:",
        row[
            "schema"
        ][
            "top_level_type"
        ]
    )

    print(
        "Top-level size:",
        row[
            "schema"
        ][
            "top_level_size"
        ]
    )

    print(
        "Legacy hits:",
        row[
            "legacy_record_hits"
        ]
    )

    print(
        "Distinct legacy IDs:",
        row[
            "distinct_legacy_case_id_count"
        ]
    )

    print(
        "Record key intersection:",
        row[
            "schema"
        ][
            "record_key_intersection"
        ]
    )

    print(
        "Record key union:",
        row[
            "schema"
        ][
            "record_key_union"
        ]
    )

    print(
        "ID-like fields:",
        cell47_registry_id_field_inventory[
            variable_name
        ]
    )

    print(
        "Planned repair action:",
        repair_action
    )


print(
    "\n=== DURABLE SHARED OBJECTS ==="
)

if (
    cell47_durable_shared_object_rows
):

    for row in (
        cell47_durable_shared_object_rows
    ):

        print(
            "-",
            row[
                "variables"
            ]
        )

else:

    print(
        "NONE"
    )


print(
    "\n=== HISTORICAL / AUDIT STRUCTURES TO PRESERVE ==="
)

for variable_name in (
    cell47_present_historical_audit_variables
):

    print(
        "-",
        variable_name
    )


print(
    "\n=== EPHEMERAL / LAST-LOOP ALIAS CANDIDATES ==="
)

for row in (
    cell47_ephemeral_alias_candidates
):

    print(
        "-",
        row[
            "variable_names"
        ],
        "| type:",
        row[
            "object_type"
        ],
        "| legacy IDs:",
        row[
            "distinct_legacy_case_id_count"
        ],
    )


print(
    "\n=== CASE -> TASK / TARGET CHAIN COVERAGE ==="
)

print(
    "Legacy cases with Task ID:",
    len(
        cell47_cases_with_task_ids
    ),
    "/ 38"
)

print(
    "Legacy cases with Target ID:",
    len(
        cell47_cases_with_target_ids
    ),
    "/ 38"
)

print(
    "Legacy cases without Task ID:",
    len(
        cell47_cases_without_task_ids
    )
)

print(
    "Legacy cases without Target ID:",
    len(
        cell47_cases_without_target_ids
    )
)


print(
    "\n=== TASK ID MULTIPLICITY ==="
)

task_id_multiplicity_counts = Counter(
    len(
        cell47_task_draft_ids_by_case[
            legacy_case_id
        ]
    )

    for legacy_case_id
    in cell46_legacy_exercise_case_ids
)


for count, cases in sorted(
    task_id_multiplicity_counts.items()
):

    print(
        "- Task IDs per legacy case:",
        count,
        "| cases:",
        cases,
    )


print(
    "\n=== TARGET ID MULTIPLICITY ==="
)

target_id_multiplicity_counts = Counter(
    len(
        cell47_target_ids_by_case[
            legacy_case_id
        ]
    )

    for legacy_case_id
    in cell46_legacy_exercise_case_ids
)


for count, cases in sorted(
    target_id_multiplicity_counts.items()
):

    print(
        "- Target IDs per legacy case:",
        count,
        "| cases:",
        cases,
    )


print(
    "\n=== PILOT DEPENDENCY ==="
)

print(
    "Legacy exercise cases represented in Pilot chain:",
    len(
        cell47_pilot_legacy_case_ids
    )
)

print(
    "Pilot legacy case IDs:",
    sorted(
        cell47_pilot_legacy_case_ids
    )
)


print(
    "\n=== REPAIR SCOPE ==="
)

for row in (
    cell47_repair_scope_rows
):

    print(
        "-",
        row[
            "variable_name"
        ],
        "=>",
        row[
            "planned_action"
        ],
    )


print(
    "\n=== CODE CELL 47 GATES ==="
)

for name, passed in (
    cell_47_gates.items()
):

    print(
        name,
        ":",
        "PASS"
        if passed
        else "FAIL"
    )


print(
    "\nCODE CELL 47 GATE:",
    "PASS"
    if cell_47_pass
    else "FAIL"
)


assert cell_47_pass, (
    "Canonical downstream dependency audit failed. "
    "Do not materialize the rebuilt Exercise cases until "
    "the actual Case -> Task -> Target -> Pilot dependency "
    "chain and alias relationships are independently resolved."
)
```

به‌خصوص ساختار و معنای این Fieldها:

این Cell تمام 38 Legacy Exercise Case را بررسی می‌کند و مشخص می‌کند:

<ul>

<li>ساختار داخلی Dict/List Fieldها چیست؛</li>
<li>آیا <code>candidate_identity_sha256</code> مستقیماً از

<li>آیا Formula فعلی <code>golden_case_id</code> از Evidence موجود قابل کشف است یا نه؛</li>
<li>کدام Fieldها برای Case جدید باید Recompute شوند و کدام‌ها قابل Preserve هستند.</li>

<p>
اگر Formula یا Semantic Contract یک Field از Evidence فعلی قابل اثبات نباشد،
وضعیت آن صریحاً
<code>NOT_YET_RESOLVED</code>
باقی می‌ماند.
</p>

<p>
در این مرحله هیچ Golden Case جدید ساخته نمی‌شود و هیچ Registry تغییر نمی‌کند.
</p>

</div>

```python
# Code Cell 48

# ============================================================
# Golden Case Materialization Contract Discovery
#
# Goal:
#   inspect the exact current 38 legacy Exercise Golden Cases
#   before creating any new Exercise-unit Golden Cases.
#
# READ-ONLY.
#
# NO case materialization.
# NO supersession.
# NO Task rebuild.
# NO Target rebuild.
# NO Pilot rebuild.
# NO question generation.
# NO artifact writes.
# ============================================================


# ============================================================
# 1) Exact 38 legacy Exercise case records
# ============================================================

cell48_legacy_exercise_case_records = [
    record
    for record
    in golden_case_records
    if (
        record[
            "golden_case_id"
        ]
        in
        cell46_legacy_exercise_case_ids
    )
]


cell48_legacy_exercise_case_records = sorted(
    cell48_legacy_exercise_case_records,
    key=lambda record:
        record[
            "golden_case_id"
        ],
)


assert (
    len(
        cell48_legacy_exercise_case_records
    )
    == 38
)


# ============================================================
# 2) Verify exact top-level schema
# ============================================================

cell48_top_level_key_sets = [
    set(
        record.keys()
    )
    for record
    in cell48_legacy_exercise_case_records
]


cell48_top_level_key_union = set().union(
    *cell48_top_level_key_sets
)


cell48_top_level_key_intersection = set(
    cell48_top_level_key_sets[
        0
    ]
)


for key_set in (
    cell48_top_level_key_sets[
        1:
    ]
):

    cell48_top_level_key_intersection &= (
        key_set
    )


cell48_top_level_schema_uniform = (
    len(
        {
            tuple(
                sorted(
                    key_set
                )
            )
            for key_set
            in cell48_top_level_key_sets
        }
    )
    == 1
)


# ============================================================
# 3) Generic value signature helper
#
# Used only for descriptive schema analysis.
# ============================================================

def cell48_value_signature(
    value,
):

    if isinstance(
        value,
        dict,
    ):

        return (
            "dict",
            tuple(
                sorted(
                    value.keys()
                )
            ),
        )


    if isinstance(
        value,
        list,
    ):

        element_types = tuple(
            sorted(
                {
                    type(
                        item
                    ).__name__
                    for item
                    in value
                }
            )
        )


        return (
            "list",
            element_types,
        )


    if isinstance(
        value,
        tuple,
    ):

        return (
            "tuple",
            tuple(
                type(
                    item
                ).__name__
                for item
                in value
            ),
        )


    return (
        type(
            value
        ).__name__,
    )


# ============================================================
# 4) Field-level value / signature distributions
# ============================================================

cell48_field_signature_counts = {}

cell48_field_distinct_value_counts = {}

cell48_field_constant_values = {}

cell48_field_example_values = {}


for field_name in sorted(
    cell48_top_level_key_union
):

    values = [
        record.get(
            field_name
        )
        for record
        in cell48_legacy_exercise_case_records
    ]


    signature_counter = Counter(
        cell48_value_signature(
            value
        )
        for value
        in values
    )


    cell48_field_signature_counts[
        field_name
    ] = (
        signature_counter
    )


    serialized_values = [
        stable_json_key(
            value
        )
        for value
        in values
    ]


    distinct_serialized_values = sorted(
        set(
            serialized_values
        )
    )


    cell48_field_distinct_value_counts[
        field_name
    ] = len(
        distinct_serialized_values
    )


    if (
        len(
            distinct_serialized_values
        )
        == 1
    ):

        cell48_field_constant_values[
            field_name
        ] = (
            values[
                0
            ]
        )


    examples = []


    seen_serialized = set()


    for value, serialized in zip(
        values,
        serialized_values,
    ):

        if (
            serialized
            in
            seen_serialized
        ):

            continue


        seen_serialized.add(
            serialized
        )


        examples.append(
            value
        )


        if (
            len(
                examples
            )
            >= 3
        ):

            break


    cell48_field_example_values[
        field_name
    ] = (
        examples
    )


# ============================================================
# 5) Deep schema inspection for complex fields
# ============================================================

CELL48_COMPLEX_FIELDS = [
    "candidate_identity",
    "candidate_snapshot",
    "evidence_references",
    "evidence_unit",
    "selection_provenance",
]


cell48_complex_field_schema = {}


for field_name in (
    CELL48_COMPLEX_FIELDS
):

    values = [
        record[
            field_name
        ]
        for record
        in cell48_legacy_exercise_case_records
    ]


    type_counts = Counter(
        type(
            value
        ).__name__
        for value
        in values
    )


    field_schema = {
        "type_counts":
            dict(
                type_counts
            ),

        "dict_key_union":
            [],

        "dict_key_intersection":
            [],

        "list_length_distribution":
            {},

        "list_element_type_counts":
            {},
    }


    dict_values = [
        value
        for value
        in values
        if isinstance(
            value,
            dict,
        )
    ]


    if (
        dict_values
    ):

        dict_key_sets = [
            set(
                value.keys()
            )
            for value
            in dict_values
        ]


        dict_key_union = set().union(
            *dict_key_sets
        )


        dict_key_intersection = set(
            dict_key_sets[
                0
            ]
        )


        for key_set in (
            dict_key_sets[
                1:
            ]
        ):

            dict_key_intersection &= (
                key_set
            )


        field_schema[
            "dict_key_union"
        ] = sorted(
            dict_key_union
        )


        field_schema[
            "dict_key_intersection"
        ] = sorted(
            dict_key_intersection
        )


    list_values = [
        value
        for value
        in values
        if isinstance(
            value,
            list,
        )
    ]


    if (
        list_values
    ):

        field_schema[
            "list_length_distribution"
        ] = dict(
            sorted(
                Counter(
                    len(
                        value
                    )
                    for value
                    in list_values
                ).items()
            )
        )


        element_type_counts = Counter()


        for value in (
            list_values
        ):

            for item in (
                value
            ):

                element_type_counts[
                    type(
                        item
                    ).__name__
                ] += 1


        field_schema[
            "list_element_type_counts"
        ] = dict(
            sorted(
                element_type_counts.items()
            )
        )


    cell48_complex_field_schema[
        field_name
    ] = (
        field_schema
    )


# ============================================================
# 6) Deep nested-key inventory for candidate identity /
# candidate snapshot / evidence references / provenance
# ============================================================

def cell48_nested_dict_schema(
    values,
):

    dict_values = [
        value
        for value
        in values
        if isinstance(
            value,
            dict,
        )
    ]


    if not (
        dict_values
    ):

        return {
            "dict_count":
                0,

            "key_union":
                [],

            "key_intersection":
                [],

            "value_type_by_key":
                {},
        }


    key_sets = [
        set(
            value.keys()
        )
        for value
        in dict_values
    ]


    key_union = set().union(
        *key_sets
    )


    key_intersection = set(
        key_sets[
            0
        ]
    )


    for key_set in (
        key_sets[
            1:
        ]
    ):

        key_intersection &= (
            key_set
        )


    value_types = defaultdict(
        set
    )


    for value in (
        dict_values
    ):

        for key, nested_value in (
            value.items()
        ):

            value_types[
                key
            ].add(
                type(
                    nested_value
                ).__name__
            )


    return {
        "dict_count":
            len(
                dict_values
            ),

        "key_union":
            sorted(
                key_union
            ),

        "key_intersection":
            sorted(
                key_intersection
            ),

        "value_type_by_key":
            {
                key:
                    sorted(
                        types
                    )

                for key, types
                in sorted(
                    value_types.items()
                )
            },
    }


cell48_nested_schema_by_complex_field = {
    field_name:
        cell48_nested_dict_schema(
            [
                record[
                    field_name
                ]
                for record
                in cell48_legacy_exercise_case_records
            ]
        )

    for field_name
    in CELL48_COMPLEX_FIELDS
}


# ============================================================
# 7) Candidate identity hash reproducibility
#
# Test ONLY the direct documented-style interpretation:
#
# SHA256(stable_json_key(candidate_identity))
#
# If it does not reproduce the stored hash, we do NOT try
# arbitrary alternative recipes here.
# ============================================================

cell48_candidate_identity_hash_matches = []

cell48_candidate_identity_hash_failures = []


for record in (
    cell48_legacy_exercise_case_records
):

    recomputed_hash = hashlib.sha256(
        stable_json_key(
            record[
                "candidate_identity"
            ]
        ).encode(
            "utf-8"
        )
    ).hexdigest()


    matches = (
        recomputed_hash
        ==
        record[
            "candidate_identity_sha256"
        ]
    )


    cell48_candidate_identity_hash_matches.append(
        matches
    )


    if not matches:

        cell48_candidate_identity_hash_failures.append(
            {
                "golden_case_id":
                    record[
                        "golden_case_id"
                    ],

                "stored_candidate_identity_sha256":
                    record[
                        "candidate_identity_sha256"
                    ],

                "direct_recomputed_sha256":
                    recomputed_hash,
            }
        )


cell48_candidate_identity_direct_hash_rule_status = (
    "REPRODUCED_FOR_ALL_38"
    if all(
        cell48_candidate_identity_hash_matches
    )
    else
    "NOT_REPRODUCED_FOR_ALL_38"
)


# ============================================================
# 8) Golden Case ID relationship audit
#
# We test only narrow evidence-supported direct relationships.
# We do NOT brute-force or invent formulas.
# ============================================================

cell48_golden_case_id_relationship_rows = []


for record in (
    cell48_legacy_exercise_case_records
):

    golden_case_id = (
        record[
            "golden_case_id"
        ]
    )


    golden_case_suffix = (
        golden_case_id.split(
            ":",
            1,
        )[
            1
        ]
        if (
            ":"
            in
            golden_case_id
        )
        else golden_case_id
    )


    candidate_hash = (
        record[
            "candidate_identity_sha256"
        ]
    )


    direct_prefix_match_32 = (
        golden_case_suffix
        ==
        candidate_hash[
            :32
        ]
    )


    direct_suffix_match_32 = (
        golden_case_suffix
        ==
        candidate_hash[
            -32:
        ]
    )


    cell48_golden_case_id_relationship_rows.append(
        {
            "golden_case_id":
                golden_case_id,

            "candidate_identity_sha256":
                candidate_hash,

            "golden_case_suffix":
                golden_case_suffix,

            "matches_candidate_hash_prefix_32":
                direct_prefix_match_32,

            "matches_candidate_hash_suffix_32":
                direct_suffix_match_32,
        }
    )


cell48_case_id_prefix32_match_count = sum(
    1
    for row
    in cell48_golden_case_id_relationship_rows
    if (
        row[
            "matches_candidate_hash_prefix_32"
        ]
    )
)


cell48_case_id_suffix32_match_count = sum(
    1
    for row
    in cell48_golden_case_id_relationship_rows
    if (
        row[
            "matches_candidate_hash_suffix_32"
        ]
    )
)


if (
    cell48_case_id_prefix32_match_count
    == 38
):

    cell48_golden_case_id_formula_status = (
        "DIRECT_CANDIDATE_IDENTITY_HASH_PREFIX_32"
    )


elif (
    cell48_case_id_suffix32_match_count
    == 38
):

    cell48_golden_case_id_formula_status = (
        "DIRECT_CANDIDATE_IDENTITY_HASH_SUFFIX_32"
    )


else:

    cell48_golden_case_id_formula_status = (
        "NOT_YET_RESOLVED"
    )


# ============================================================
# 9) Evidence document identity consistency
# ============================================================

cell48_evidence_consistency_rows = []


for record in (
    cell48_legacy_exercise_case_records
):

    evidence_document_ids = list(
        record[
            "evidence_document_ids"
        ]
    )


    evidence_document_count = (
        record[
            "evidence_document_count"
        ]
    )


    evidence_ids_exist = all(
        document_id
        in
        corpus_target_index

        for document_id
        in evidence_document_ids
    )


    count_matches = (
        evidence_document_count
        ==
        len(
            evidence_document_ids
        )
    )


    cell48_evidence_consistency_rows.append(
        {
            "golden_case_id":
                record[
                    "golden_case_id"
                ],

            "evidence_document_count":
                evidence_document_count,

            "evidence_document_ids_length":
                len(
                    evidence_document_ids
                ),

            "count_matches":
                count_matches,

            "all_documents_exist":
                evidence_ids_exist,
        }
    )


# ============================================================
# 10) Inspect selection rank behavior
# ============================================================

cell48_selection_rank_values = [
    record[
        "selection_rank"
    ]
    for record
    in cell48_legacy_exercise_case_records
]


cell48_selection_rank_distribution = Counter(
    cell48_selection_rank_values
)


cell48_selection_rank_unique_count = len(
    set(
        cell48_selection_rank_values
    )
)


# ============================================================
# 11) Determine field materialization policy from evidence
#
# This is a design classification only.
# No new records are built.
# ============================================================

CELL48_ALWAYS_RECOMPUTE_FIELDS = {
    "golden_case_id",
    "candidate_identity",
    "candidate_identity_sha256",
    "candidate_snapshot",
    "evidence_document_count",
    "evidence_document_ids",
    "evidence_references",
    "evidence_unit",
    "selection_provenance",
    "selection_rank",
}


CELL48_STATUS_FIELDS = {
    "expected_answer_generation_status",
    "final_golden_membership",
    "membership_status",
    "question_generation_status",
    "review_status",
}


cell48_field_materialization_policy = {}


for field_name in sorted(
    cell48_top_level_key_union
):

    if (
        field_name
        in
        CELL48_ALWAYS_RECOMPUTE_FIELDS
    ):

        policy = (
            "RECOMPUTE_FROM_REBUILT_EXERCISE_UNIT"
        )


    elif (
        field_name
        in
        CELL48_STATUS_FIELDS
    ):

        if (
            field_name
            in
            cell48_field_constant_values
        ):

            policy = (
                "PRESERVE_CONSTANT_STATUS_IF_SEMANTICALLY_STILL_VALID"
            )

        else:

            policy = (
                "RECOMPUTE_OR_REVIEW_STATUS"
            )


    elif (
        field_name
        in
        cell48_field_constant_values
    ):

        policy = (
            "PRESERVE_CONSTANT_VALUE"
        )


    else:

        policy = (
            "INSPECT_BEFORE_MATERIALIZATION"
        )


    cell48_field_materialization_policy[
        field_name
    ] = (
        policy
    )


# ============================================================
# 12) Identify any unresolved fields
# ============================================================

cell48_unresolved_materialization_fields = sorted(
    [
        field_name

        for field_name, policy
        in cell48_field_materialization_policy.items()

        if (
            policy
            in {
                "INSPECT_BEFORE_MATERIALIZATION",
                "RECOMPUTE_OR_REVIEW_STATUS",
            }
        )
    ]
)


# ============================================================
# 13) Record-level compact examples
#
# Select deterministic examples:
# - lowest Golden Case ID
# - highest Golden Case ID
# - failed Exercise Pilot case
#
# No semantic sampling.
# ============================================================

cell48_example_case_ids = {
    cell48_legacy_exercise_case_records[
        0
    ][
        "golden_case_id"
    ],

    cell48_legacy_exercise_case_records[
        -1
    ][
        "golden_case_id"
    ],

    CELL46_FAILED_EXERCISE_PILOT_LEGACY_CASE_ID,
}


cell48_example_case_records = [
    record
    for record
    in cell48_legacy_exercise_case_records
    if (
        record[
            "golden_case_id"
        ]
        in
        cell48_example_case_ids
    )
]


# ============================================================
# 14) Audit Registry
# ============================================================

GOLDEN_CASE_MATERIALIZATION_CONTRACT_DISCOVERY_V1_DRAFT = {
    "version":
        "golden_case_materialization_contract_discovery_v1_draft",

    "status":
        "SCHEMA_AND_DIRECT_HASH_RELATIONSHIPS_AUDITED_NO_MUTATION",

    "legacy_exercise_case_count":
        len(
            cell48_legacy_exercise_case_records
        ),

    "top_level_schema_uniform":
        cell48_top_level_schema_uniform,

    "candidate_identity_direct_hash_rule_status":
        cell48_candidate_identity_direct_hash_rule_status,

    "golden_case_id_formula_status":
        cell48_golden_case_id_formula_status,

    "unresolved_materialization_fields":
        list(
            cell48_unresolved_materialization_fields
        ),

    "new_case_materialization_status":
        "NOT_STARTED",

    "legacy_supersession_status":
        "NOT_STARTED",

    "task_rebuild_status":
        "NOT_STARTED",

    "target_rebuild_status":
        "NOT_STARTED",

    "pilot_rebuild_status":
        "NOT_STARTED",

    "question_generation_status":
        "NOT_STARTED",

    "artifact_write_status":
        "NONE",
}


# ============================================================
# 15) Mutation guardrail snapshots
# ============================================================

cell48_golden_case_records_count_before = (
    len(
        golden_case_records
    )
)


cell48_auto_case_records_count_before = (
    len(
        auto_case_records
    )
)


cell48_selected_sample_before = set(
    selected_exercise_population_unit_ids
)


# ============================================================
# 16) Validation gates
# ============================================================

exact_38_case_records_gate = (
    len(
        cell48_legacy_exercise_case_records
    )
    == 38
)


uniform_schema_gate = (
    cell48_top_level_schema_uniform
    is True

    and

    cell48_top_level_key_union
    ==
    cell48_top_level_key_intersection
)


candidate_identity_hash_audit_completed_gate = (
    len(
        cell48_candidate_identity_hash_matches
    )
    == 38
)


golden_case_id_relationship_audit_completed_gate = (
    len(
        cell48_golden_case_id_relationship_rows
    )
    == 38
)


evidence_counts_consistent_gate = all(
    row[
        "count_matches"
    ]

    and

    row[
        "all_documents_exist"
    ]

    for row
    in cell48_evidence_consistency_rows
)


all_fields_have_materialization_policy_gate = (
    set(
        cell48_field_materialization_policy
    )
    ==
    cell48_top_level_key_union
)


no_silent_case_id_formula_guess_gate = (
    cell48_golden_case_id_formula_status
    in {
        "DIRECT_CANDIDATE_IDENTITY_HASH_PREFIX_32",
        "DIRECT_CANDIDATE_IDENTITY_HASH_SUFFIX_32",
        "NOT_YET_RESOLVED",
    }
)


golden_records_not_mutated_gate = (
    len(
        golden_case_records
    )
    ==
    cell48_golden_case_records_count_before
)


auto_records_not_mutated_gate = (
    len(
        auto_case_records
    )
    ==
    cell48_auto_case_records_count_before
)


sample_not_mutated_gate = (
    set(
        selected_exercise_population_unit_ids
    )
    ==
    cell48_selected_sample_before
)


case_materialization_not_started_gate = (
    GOLDEN_CASE_MATERIALIZATION_CONTRACT_DISCOVERY_V1_DRAFT[
        "new_case_materialization_status"
    ]
    ==
    "NOT_STARTED"
)


supersession_not_started_gate = (
    GOLDEN_CASE_MATERIALIZATION_CONTRACT_DISCOVERY_V1_DRAFT[
        "legacy_supersession_status"
    ]
    ==
    "NOT_STARTED"
)


task_rebuild_not_started_gate = (
    GOLDEN_CASE_MATERIALIZATION_CONTRACT_DISCOVERY_V1_DRAFT[
        "task_rebuild_status"
    ]
    ==
    "NOT_STARTED"
)


target_rebuild_not_started_gate = (
    GOLDEN_CASE_MATERIALIZATION_CONTRACT_DISCOVERY_V1_DRAFT[
        "target_rebuild_status"
    ]
    ==
    "NOT_STARTED"
)


pilot_rebuild_not_started_gate = (
    GOLDEN_CASE_MATERIALIZATION_CONTRACT_DISCOVERY_V1_DRAFT[
        "pilot_rebuild_status"
    ]
    ==
    "NOT_STARTED"
)


question_generation_not_started_gate = (
    GOLDEN_CASE_MATERIALIZATION_CONTRACT_DISCOVERY_V1_DRAFT[
        "question_generation_status"
    ]
    ==
    "NOT_STARTED"
)


no_artifact_write_gate = (
    GOLDEN_CASE_MATERIALIZATION_CONTRACT_DISCOVERY_V1_DRAFT[
        "artifact_write_status"
    ]
    ==
    "NONE"
)


cell_48_gates = {
    "exact_38_legacy_exercise_case_records_inspected":
        exact_38_case_records_gate,

    "legacy_exercise_case_top_level_schema_is_uniform":
        uniform_schema_gate,

    "candidate_identity_sha256_direct_rule_audited_for_all_38":
        candidate_identity_hash_audit_completed_gate,

    "golden_case_id_direct_relationship_audited_for_all_38":
        golden_case_id_relationship_audit_completed_gate,

    "all_legacy_evidence_document_counts_and_ids_are_consistent":
        evidence_counts_consistent_gate,

    "every_golden_case_field_has_explicit_materialization_policy":
        all_fields_have_materialization_policy_gate,

    "golden_case_id_formula_not_silently_guessed":
        no_silent_case_id_formula_guess_gate,

    "golden_case_records_not_mutated":
        golden_records_not_mutated_gate,

    "auto_case_records_not_mutated":
        auto_records_not_mutated_gate,

    "rebuilt_exercise_sample_not_mutated":
        sample_not_mutated_gate,

    "new_case_materialization_not_started":
        case_materialization_not_started_gate,

    "legacy_supersession_not_started":
        supersession_not_started_gate,

    "task_rebuild_not_started":
        task_rebuild_not_started_gate,

    "target_rebuild_not_started":
        target_rebuild_not_started_gate,

    "pilot_rebuild_not_started":
        pilot_rebuild_not_started_gate,

    "question_generation_not_started":
        question_generation_not_started_gate,

    "no_artifact_write":
        no_artifact_write_gate,
}


cell_48_pass = all(
    cell_48_gates.values()
)


# ============================================================
# 17) Output
# ============================================================

print(
    "=== GOLDEN CASE MATERIALIZATION CONTRACT DISCOVERY ==="
)

print(
    "Legacy Exercise cases inspected:",
    len(
        cell48_legacy_exercise_case_records
    )
)

print(
    "Uniform top-level schema:",
    cell48_top_level_schema_uniform
)

print(
    "Top-level fields:",
    sorted(
        cell48_top_level_key_union
    )
)


print(
    "\n=== FIELD DISTINCT-VALUE COUNTS ==="
)

for field_name in sorted(
    cell48_field_distinct_value_counts
):

    print(
        "-",
        field_name,
        ":",
        cell48_field_distinct_value_counts[
            field_name
        ],
    )


print(
    "\n=== CONSTANT FIELDS ==="
)

for field_name, value in sorted(
    cell48_field_constant_values.items()
):

    print(
        "-",
        field_name,
        "=",
        value,
    )


print(
    "\n=== COMPLEX FIELD SCHEMA ==="
)

for field_name in (
    CELL48_COMPLEX_FIELDS
):

    print(
        "\nFIELD:",
        field_name
    )

    print(
        "Type counts:",
        cell48_complex_field_schema[
            field_name
        ][
            "type_counts"
        ]
    )

    print(
        "Nested key intersection:",
        cell48_nested_schema_by_complex_field[
            field_name
        ][
            "key_intersection"
        ]
    )

    print(
        "Nested key union:",
        cell48_nested_schema_by_complex_field[
            field_name
        ][
            "key_union"
        ]
    )

    print(
        "Nested value types:",
        cell48_nested_schema_by_complex_field[
            field_name
        ][
            "value_type_by_key"
        ]
    )

    print(
        "List-length distribution:",
        cell48_complex_field_schema[
            field_name
        ][
            "list_length_distribution"
        ]
    )


print(
    "\n=== CANDIDATE IDENTITY HASH AUDIT ==="
)

print(
    "Direct SHA256(stable_json_key(candidate_identity)) status:",
    cell48_candidate_identity_direct_hash_rule_status
)

print(
    "Failures:",
    len(
        cell48_candidate_identity_hash_failures
    )
)


if (
    cell48_candidate_identity_hash_failures
):

    for failure in (
        cell48_candidate_identity_hash_failures[
            :5
        ]
    ):

        print(
            "-",
            failure
        )


print(
    "\n=== GOLDEN CASE ID RELATIONSHIP AUDIT ==="
)

print(
    "Candidate-hash prefix-32 matches:",
    cell48_case_id_prefix32_match_count,
    "/ 38"
)

print(
    "Candidate-hash suffix-32 matches:",
    cell48_case_id_suffix32_match_count,
    "/ 38"
)

print(
    "Formula status:",
    cell48_golden_case_id_formula_status
)


print(
    "\n=== SELECTION RANK AUDIT ==="
)

print(
    "Unique selection-rank values:",
    cell48_selection_rank_unique_count
)

print(
    "Selection-rank distribution:",
    dict(
        sorted(
            cell48_selection_rank_distribution.items(),
            key=lambda item:
                str(
                    item[
                        0
                    ]
                ),
        )
    )
)


print(
    "\n=== FIELD MATERIALIZATION POLICY ==="
)

for field_name, policy in sorted(
    cell48_field_materialization_policy.items()
):

    print(
        "-",
        field_name,
        "=>",
        policy,
    )


print(
    "\nUnresolved materialization fields:",
    cell48_unresolved_materialization_fields
)


print(
    "\n=== DETERMINISTIC EXAMPLE CASE RECORDS ==="
)

for record in (
    cell48_example_case_records
):

    print(
        "\nCASE:",
        record[
            "golden_case_id"
        ]
    )

    for field_name in sorted(
        record
    ):

        print(
            field_name,
            ":",
            record[
                field_name
            ],
        )


print(
    "\n=== CODE CELL 48 GATES ==="
)

for name, passed in (
    cell_48_gates.items()
):

    print(
        name,
        ":",
        "PASS"
        if passed
        else "FAIL"
    )


print(
    "\nCODE CELL 48 GATE:",
    "PASS"
    if cell_48_pass
    else "FAIL"
)


assert cell_48_pass, (
    "Golden Case materialization contract discovery failed. "
    "Do not construct the rebuilt 38 Exercise Golden Cases "
    "until the current case schema, identity hash behavior, "
    "evidence fields and materialization policies are fully "
    "audited."
)
```

```python
# Code Cell 49 — REPAIRED

# ============================================================
# Golden Case ID Formula Discovery
#
# REPAIR:
#   Never iterate directly over globals().items() at notebook
#   top level because loop-variable creation can mutate the
#   global namespace during iteration.
#
#   A fixed namespace snapshot is created once and all
#   discovery scans use that immutable key/value snapshot.
#
# Goal:
#   discover the actual legacy golden_case_id construction
#   formula from the executed notebook / current kernel.
#
# READ-ONLY.
#
# NO formula guessing.
# NO new case materialization.
# NO supersession.
# NO Task rebuild.
# NO Target rebuild.
# NO Pilot rebuild.
# NO question generation.
# NO artifact writes.
# ============================================================


# ============================================================
# 0) Freeze current namespace for safe inspection
#
# RHS is fully evaluated before cell49_globals_snapshot is
# assigned into the global namespace.
# ============================================================

cell49_globals_snapshot = dict(
    globals()
)


cell49_globals_snapshot_items = list(
    cell49_globals_snapshot.items()
)


# ============================================================
# 1) Locate the current notebook file from already-established
# runtime variables / repo structure.
#
# No path is guessed silently.
# ============================================================

cell49_notebook_path_candidates = []


for variable_name, value in (
    cell49_globals_snapshot_items
):

    if isinstance(
        value,
        Path,
    ):

        value_string = str(
            value
        )


        if (
            value_string.endswith(
                "12_golden_dataset.ipynb"
            )
        ):

            cell49_notebook_path_candidates.append(
                {
                    "source_variable":
                        variable_name,

                    "path":
                        value,
                }
            )


# ------------------------------------------------------------
# Fallback only within already-established repository roots.
# ------------------------------------------------------------

if not (
    cell49_notebook_path_candidates
):

    possible_repo_roots = []


    for variable_name, value in (
        cell49_globals_snapshot_items
    ):

        if isinstance(
            value,
            Path,
        ):

            phase2_dir_candidate = (
                value
                /
                "notebooks"
                /
                "phase2"
            )


            if (
                phase2_dir_candidate.exists()
            ):

                possible_repo_roots.append(
                    {
                        "source_variable":
                            variable_name,

                        "path":
                            value,
                    }
                )


    unique_repo_root_paths = {
        str(
            row[
                "path"
            ].resolve()
        ):
            row

        for row
        in possible_repo_roots
    }


    for row in (
        unique_repo_root_paths.values()
    ):

        candidate_path = (
            row[
                "path"
            ]
            /
            "notebooks"
            /
            "phase2"
            /
            "12_golden_dataset.ipynb"
        )


        if (
            candidate_path.exists()
        ):

            cell49_notebook_path_candidates.append(
                {
                    "source_variable":
                        (
                            "derived_from:"
                            +
                            row[
                                "source_variable"
                            ]
                        ),

                    "path":
                        candidate_path,
                }
            )


# ------------------------------------------------------------
# Deduplicate exact resolved paths.
# ------------------------------------------------------------

cell49_notebook_path_candidates = list(
    {
        str(
            row[
                "path"
            ].resolve()
        ):
            {
                **row,

                "path":
                    row[
                        "path"
                    ].resolve(),
            }

        for row
        in cell49_notebook_path_candidates
    }.values()
)


assert (
    len(
        cell49_notebook_path_candidates
    )
    == 1
), (
    "Expected exactly one grounded NB12 notebook path; "
    f"found {len(cell49_notebook_path_candidates)}."
)


cell49_notebook_path = (
    cell49_notebook_path_candidates[
        0
    ][
        "path"
    ]
)


# ============================================================
# 2) Read actual notebook JSON
# ============================================================

with cell49_notebook_path.open(
    "r",
    encoding="utf-8",
) as handle:

    cell49_notebook_json = json.load(
        handle
    )


cell49_code_cells = [
    {
        "cell_index":
            index,

        "source":
            (
                cell[
                    "source"
                ]
                if isinstance(
                    cell[
                        "source"
                    ],
                    str,
                )
                else
                "".join(
                    cell[
                        "source"
                    ]
                )
            ),
    }

    for index, cell
    in enumerate(
        cell49_notebook_json.get(
            "cells",
            []
        )
    )

    if (
        cell.get(
            "cell_type"
        )
        ==
        "code"
    )
]


# ============================================================
# 3) Search actual notebook code for Golden Case ID builders
# ============================================================

CELL49_CODE_SEARCH_TERMS = [
    "golden_case_id",
    "golden-case:",
    "candidate_identity_sha256",
    "candidate_identity",
    "stable_json_key",
]


cell49_candidate_source_cells = []


for code_cell in (
    cell49_code_cells
):

    source = (
        code_cell[
            "source"
        ]
    )


    matched_terms = [
        term

        for term
        in CELL49_CODE_SEARCH_TERMS

        if (
            term
            in
            source
        )
    ]


    if not (
        (
            "golden_case_id"
            in
            matched_terms
        )

        or

        (
            "golden-case:"
            in
            matched_terms
        )
    ):

        continue


    cell49_candidate_source_cells.append(
        {
            "cell_index":
                code_cell[
                    "cell_index"
                ],

            "matched_terms":
                matched_terms,

            "source":
                source,
        }
    )


# ============================================================
# 4) Search FIXED kernel snapshot for callables likely related
# to Case-ID construction.
#
# No callable is executed yet.
# ============================================================

cell49_candidate_callable_rows = []


for variable_name, value in sorted(
    cell49_globals_snapshot_items,
    key=lambda item:
        item[
            0
        ],
):

    if not callable(
        value
    ):

        continue


    lowered_name = (
        variable_name.lower()
    )


    if not (
        (
            "golden"
            in
            lowered_name
        )

        or

        (
            "case"
            in
            lowered_name

            and

            "id"
            in
            lowered_name
        )

        or

        (
            "stable"
            in
            lowered_name

            and

            "id"
            in
            lowered_name
        )
    ):

        continue


    code_object = getattr(
        value,
        "__code__",
        None,
    )


    if (
        code_object
        is None
    ):

        continue


    cell49_candidate_callable_rows.append(
        {
            "name":
                variable_name,

            "arg_names":
                list(
                    code_object.co_varnames[
                        :code_object.co_argcount
                    ]
                ),

            "referenced_names":
                list(
                    code_object.co_names
                ),

            "constants":
                [
                    constant

                    for constant
                    in code_object.co_consts

                    if isinstance(
                        constant,
                        (
                            str,
                            int,
                            float,
                            bool,
                            type(
                                None
                            ),
                        ),
                    )
                ],
        }
    )


# ============================================================
# 5) Extract exact source snippets around golden_case_id
# construction.
# ============================================================

cell49_formula_evidence_snippets = []


for row in (
    cell49_candidate_source_cells
):

    source_lines = (
        row[
            "source"
        ].splitlines()
    )


    matched_line_indexes = [
        index

        for index, line
        in enumerate(
            source_lines
        )

        if (
            "golden_case_id"
            in
            line

            or

            "golden-case:"
            in
            line
        )
    ]


    for matched_index in (
        matched_line_indexes
    ):

        start_index = max(
            0,
            matched_index
            - 8,
        )


        end_index = min(
            len(
                source_lines
            ),
            matched_index
            + 12,
        )


        snippet = "\n".join(
            source_lines[
                start_index:
                end_index
            ]
        )


        cell49_formula_evidence_snippets.append(
            {
                "cell_index":
                    row[
                        "cell_index"
                    ],

                "matched_line_index":
                    matched_index,

                "snippet":
                    snippet,
            }
        )


# ============================================================
# 6) Discover existing Case-ID builder functions by safe
# behavioral validation.
#
# Only candidate callables from the PRE-CELL snapshot are
# considered.
# ============================================================

cell49_safe_builder_candidates = []


for row in (
    cell49_candidate_callable_rows
):

    callable_name = (
        row[
            "name"
        ]
    )


    lowered_name = (
        callable_name.lower()
    )


    if not (
        "case"
        in
        lowered_name

        and

        "id"
        in
        lowered_name
    ):

        continue


    function_object = (
        cell49_globals_snapshot.get(
            callable_name
        )
    )


    if (
        function_object
        is None
    ):

        continue


    arg_names = (
        row[
            "arg_names"
        ]
    )


    supported_argument_names = {
        "candidate_identity",
        "candidate_identity_sha256",
        "source_type",
        "stratum",
        "selection_rank",
        "candidate_snapshot",
        "evidence_document_ids",
        "case_schema_version",
    }


    if not (
        set(
            arg_names
        )
        <=
        supported_argument_names
    ):

        continue


    cell49_safe_builder_candidates.append(
        {
            **row,

            "callable":
                function_object,
        }
    )


# ============================================================
# 7) Validate each safe candidate against all 38 existing
# legacy Exercise Golden Cases.
#
# Accepted only with exact 38 / 38 reproduction.
# ============================================================

cell49_builder_validation_rows = []


for candidate in (
    cell49_safe_builder_candidates
):

    function_object = (
        candidate[
            "callable"
        ]
    )


    arg_names = (
        candidate[
            "arg_names"
        ]
    )


    matches = 0

    errors = []

    mismatches = []


    for record in (
        cell48_legacy_exercise_case_records
    ):

        kwargs = {
            argument_name:
                record[
                    argument_name
                ]

            for argument_name
            in arg_names
        }


        try:

            produced_value = (
                function_object(
                    **kwargs
                )
            )


        except Exception as exc:

            errors.append(
                {
                    "golden_case_id":
                        record[
                            "golden_case_id"
                        ],

                    "error_type":
                        type(
                            exc
                        ).__name__,

                    "error_message":
                        str(
                            exc
                        ),
                }
            )

            continue


        if (
            produced_value
            ==
            record[
                "golden_case_id"
            ]
        ):

            matches += 1


        else:

            mismatches.append(
                {
                    "stored":
                        record[
                            "golden_case_id"
                        ],

                    "produced":
                        produced_value,
                }
            )


    cell49_builder_validation_rows.append(
        {
            "builder_name":
                candidate[
                    "name"
                ],

            "arg_names":
                arg_names,

            "exact_match_count":
                matches,

            "mismatch_count":
                len(
                    mismatches
                ),

            "error_count":
                len(
                    errors
                ),

            "errors":
                errors,

            "mismatches":
                mismatches,

            "reproduces_all_38":
                (
                    matches
                    == 38

                    and

                    len(
                        mismatches
                    )
                    == 0

                    and

                    len(
                        errors
                    )
                    == 0
                ),
        }
    )


cell49_exact_builder_matches = [
    row

    for row
    in cell49_builder_validation_rows

    if (
        row[
            "reproduces_all_38"
        ]
        is True
    )
]


# ============================================================
# 8) Conservative source-formula signals
#
# No eval and no arbitrary formula search.
# ============================================================

cell49_recognized_formula_candidates = []


for source_row in (
    cell49_candidate_source_cells
):

    source = (
        source_row[
            "source"
        ]
    )


    if (
        "golden-case:"
        in
        source

        and

        "hashlib.sha256"
        in
        source

        and

        "stable_json_key"
        in
        source
    ):

        cell49_recognized_formula_candidates.append(
            {
                "cell_index":
                    source_row[
                        "cell_index"
                    ],

                "formula_class":
                    (
                        "SHA256_STABLE_JSON_PAYLOAD_"
                        "WITH_GOLDEN_CASE_PREFIX"
                    ),

                "source":
                    source,
            }
        )


# ============================================================
# 9) Search likely payload names surrounding actual assignment
# ============================================================

CELL49_PAYLOAD_NAME_TERMS = [
    "identity_payload",
    "case_identity",
    "case_id_payload",
    "golden_case_payload",
    "case_payload",
]


cell49_payload_signal_rows = []


for snippet_row in (
    cell49_formula_evidence_snippets
):

    matched_payload_terms = [
        term

        for term
        in CELL49_PAYLOAD_NAME_TERMS

        if (
            term
            in
            snippet_row[
                "snippet"
            ]
        )
    ]


    if (
        matched_payload_terms
    ):

        cell49_payload_signal_rows.append(
            {
                **snippet_row,

                "payload_terms":
                    matched_payload_terms,
            }
        )


# ============================================================
# 10) Resolve formula status
#
# Strong evidence only:
# exactly one existing callable reproduces 38 / 38.
#
# Otherwise source evidence is printed and the Cell FAILs so
# this SAME numbered Cell can be repaired using exact evidence.
# ============================================================

if (
    len(
        cell49_exact_builder_matches
    )
    == 1
):

    cell49_golden_case_id_formula_status = (
        "RESOLVED_BY_EXISTING_BUILDER_38_OF_38"
    )


    cell49_resolved_builder_name = (
        cell49_exact_builder_matches[
            0
        ][
            "builder_name"
        ]
    )


else:

    cell49_golden_case_id_formula_status = (
        "NOT_YET_RESOLVED"
    )


    cell49_resolved_builder_name = None


# ============================================================
# 11) Independent second reproduction pass
# ============================================================

cell49_independent_reproduction_failures = []


if (
    cell49_resolved_builder_name
    is not None
):

    cell49_resolved_builder = (
        cell49_globals_snapshot[
            cell49_resolved_builder_name
        ]
    )


    cell49_resolved_builder_code = getattr(
        cell49_resolved_builder,
        "__code__",
        None,
    )


    assert (
        cell49_resolved_builder_code
        is not None
    )


    cell49_resolved_builder_args = list(
        cell49_resolved_builder_code.co_varnames[
            :cell49_resolved_builder_code.co_argcount
        ]
    )


    for record in (
        cell48_legacy_exercise_case_records
    ):

        kwargs = {
            argument_name:
                record[
                    argument_name
                ]

            for argument_name
            in cell49_resolved_builder_args
        }


        recomputed_case_id = (
            cell49_resolved_builder(
                **kwargs
            )
        )


        if (
            recomputed_case_id
            !=
            record[
                "golden_case_id"
            ]
        ):

            cell49_independent_reproduction_failures.append(
                {
                    "stored":
                        record[
                            "golden_case_id"
                        ],

                    "recomputed":
                        recomputed_case_id,
                }
            )


# ============================================================
# 12) Audit registry
# ============================================================

GOLDEN_CASE_ID_FORMULA_DISCOVERY_V1_DRAFT = {
    "version":
        "golden_case_id_formula_discovery_v1_draft",

    "status":
        cell49_golden_case_id_formula_status,

    "notebook_path":
        str(
            cell49_notebook_path
        ),

    "namespace_snapshot_size":
        len(
            cell49_globals_snapshot
        ),

    "candidate_source_cell_count":
        len(
            cell49_candidate_source_cells
        ),

    "formula_evidence_snippet_count":
        len(
            cell49_formula_evidence_snippets
        ),

    "candidate_callable_count":
        len(
            cell49_candidate_callable_rows
        ),

    "safe_builder_candidate_count":
        len(
            cell49_safe_builder_candidates
        ),

    "exact_builder_match_count":
        len(
            cell49_exact_builder_matches
        ),

    "resolved_builder_name":
        cell49_resolved_builder_name,

    "independent_reproduction_failure_count":
        len(
            cell49_independent_reproduction_failures
        ),

    "case_materialization_status":
        "NOT_STARTED",

    "legacy_supersession_status":
        "NOT_STARTED",

    "task_rebuild_status":
        "NOT_STARTED",

    "target_rebuild_status":
        "NOT_STARTED",

    "pilot_rebuild_status":
        "NOT_STARTED",

    "question_generation_status":
        "NOT_STARTED",

    "artifact_write_status":
        "NONE",
}


# ============================================================
# 13) Mutation guardrail snapshots
# ============================================================

cell49_golden_case_count_before = len(
    golden_case_records
)


cell49_task_plan_count_before = len(
    task_draft_plan_rows
)


cell49_target_count_before = len(
    target_evidence_records
)


cell49_sample_ids_before = set(
    selected_exercise_population_unit_ids
)


# ============================================================
# 14) Validation gates
#
# Formula resolution is deliberately required.
# If no exact builder is found, this Cell FAILs after printing
# source evidence. Repair this SAME Cell 49; do not advance.
# ============================================================

exactly_one_notebook_path_gate = (
    len(
        cell49_notebook_path_candidates
    )
    == 1
)


notebook_readable_gate = (
    isinstance(
        cell49_notebook_json,
        dict,
    )

    and

    len(
        cell49_code_cells
    )
    > 0
)


source_formula_evidence_found_gate = (
    len(
        cell49_candidate_source_cells
    )
    > 0

    and

    len(
        cell49_formula_evidence_snippets
    )
    > 0
)


exactly_one_validated_builder_gate = (
    len(
        cell49_exact_builder_matches
    )
    == 1
)


formula_resolved_gate = (
    cell49_golden_case_id_formula_status
    ==
    "RESOLVED_BY_EXISTING_BUILDER_38_OF_38"
)


independent_38_of_38_reproduction_gate = (
    formula_resolved_gate

    and

    len(
        cell49_independent_reproduction_failures
    )
    == 0
)


golden_cases_not_mutated_gate = (
    len(
        golden_case_records
    )
    ==
    cell49_golden_case_count_before
)


tasks_not_mutated_gate = (
    len(
        task_draft_plan_rows
    )
    ==
    cell49_task_plan_count_before
)


targets_not_mutated_gate = (
    len(
        target_evidence_records
    )
    ==
    cell49_target_count_before
)


sample_not_mutated_gate = (
    set(
        selected_exercise_population_unit_ids
    )
    ==
    cell49_sample_ids_before
)


case_materialization_not_started_gate = (
    GOLDEN_CASE_ID_FORMULA_DISCOVERY_V1_DRAFT[
        "case_materialization_status"
    ]
    ==
    "NOT_STARTED"
)


supersession_not_started_gate = (
    GOLDEN_CASE_ID_FORMULA_DISCOVERY_V1_DRAFT[
        "legacy_supersession_status"
    ]
    ==
    "NOT_STARTED"
)


task_rebuild_not_started_gate = (
    GOLDEN_CASE_ID_FORMULA_DISCOVERY_V1_DRAFT[
        "task_rebuild_status"
    ]
    ==
    "NOT_STARTED"
)


target_rebuild_not_started_gate = (
    GOLDEN_CASE_ID_FORMULA_DISCOVERY_V1_DRAFT[
        "target_rebuild_status"
    ]
    ==
    "NOT_STARTED"
)


pilot_rebuild_not_started_gate = (
    GOLDEN_CASE_ID_FORMULA_DISCOVERY_V1_DRAFT[
        "pilot_rebuild_status"
    ]
    ==
    "NOT_STARTED"
)


question_generation_not_started_gate = (
    GOLDEN_CASE_ID_FORMULA_DISCOVERY_V1_DRAFT[
        "question_generation_status"
    ]
    ==
    "NOT_STARTED"
)


no_artifact_write_gate = (
    GOLDEN_CASE_ID_FORMULA_DISCOVERY_V1_DRAFT[
        "artifact_write_status"
    ]
    ==
    "NONE"
)


cell_49_gates = {
    "exactly_one_grounded_nb12_notebook_path_found":
        exactly_one_notebook_path_gate,

    "nb12_notebook_json_read_successfully":
        notebook_readable_gate,

    "actual_notebook_source_contains_golden_case_id_formula_evidence":
        source_formula_evidence_found_gate,

    "exactly_one_existing_builder_reproduces_all_38_legacy_case_ids":
        exactly_one_validated_builder_gate,

    "golden_case_id_formula_resolved_without_guessing":
        formula_resolved_gate,

    "resolved_builder_independently_reproduces_38_of_38":
        independent_38_of_38_reproduction_gate,

    "golden_case_records_not_mutated":
        golden_cases_not_mutated_gate,

    "task_draft_records_not_mutated":
        tasks_not_mutated_gate,

    "target_evidence_records_not_mutated":
        targets_not_mutated_gate,

    "rebuilt_exercise_sample_not_mutated":
        sample_not_mutated_gate,

    "new_case_materialization_not_started":
        case_materialization_not_started_gate,

    "legacy_supersession_not_started":
        supersession_not_started_gate,

    "task_rebuild_not_started":
        task_rebuild_not_started_gate,

    "target_rebuild_not_started":
        target_rebuild_not_started_gate,

    "pilot_rebuild_not_started":
        pilot_rebuild_not_started_gate,

    "question_generation_not_started":
        question_generation_not_started_gate,

    "no_artifact_write":
        no_artifact_write_gate,
}


cell_49_pass = all(
    cell_49_gates.values()
)


# ============================================================
# 15) Output
# ============================================================

print(
    "=== GOLDEN CASE ID FORMULA DISCOVERY ==="
)

print(
    "Notebook:",
    cell49_notebook_path
)

print(
    "Grounded notebook source variable:",
    cell49_notebook_path_candidates[
        0
    ][
        "source_variable"
    ]
)

print(
    "Frozen namespace objects inspected:",
    len(
        cell49_globals_snapshot
    )
)

print(
    "Code cells inspected:",
    len(
        cell49_code_cells
    )
)

print(
    "Candidate source cells:",
    len(
        cell49_candidate_source_cells
    )
)

print(
    "Formula evidence snippets:",
    len(
        cell49_formula_evidence_snippets
    )
)


print(
    "\n=== CANDIDATE KERNEL CALLABLES ==="
)

if (
    cell49_candidate_callable_rows
):

    for row in (
        cell49_candidate_callable_rows
    ):

        print(
            "-",
            row[
                "name"
            ],
            "| args:",
            row[
                "arg_names"
            ],
            "| referenced names:",
            row[
                "referenced_names"
            ],
        )

else:

    print(
        "NONE"
    )


print(
    "\n=== SAFE CASE-ID BUILDER VALIDATION ==="
)

if (
    cell49_builder_validation_rows
):

    for row in (
        cell49_builder_validation_rows
    ):

        print(
            "-",
            row[
                "builder_name"
            ],
            "| matches:",
            row[
                "exact_match_count"
            ],
            "/ 38",
            "| mismatches:",
            row[
                "mismatch_count"
            ],
            "| errors:",
            row[
                "error_count"
            ],
            "| exact:",
            row[
                "reproduces_all_38"
            ],
        )

else:

    print(
        "NONE"
    )


print(
    "\n=== FORMULA SOURCE EVIDENCE ==="
)

for index, row in enumerate(
    cell49_formula_evidence_snippets[
        :20
    ],
    start=1,
):

    print(
        "\nSNIPPET",
        index,
        "| notebook cell index:",
        row[
            "cell_index"
        ],
        "| matched line:",
        row[
            "matched_line_index"
        ],
    )

    print(
        row[
            "snippet"
        ]
    )


print(
    "\n=== RECOGNIZED SOURCE FORMULA SIGNALS ==="
)

if (
    cell49_recognized_formula_candidates
):

    for row in (
        cell49_recognized_formula_candidates
    ):

        print(
            "- cell:",
            row[
                "cell_index"
            ],
            "| class:",
            row[
                "formula_class"
            ],
        )

else:

    print(
        "NONE"
    )


print(
    "\n=== PAYLOAD SIGNALS ==="
)

if (
    cell49_payload_signal_rows
):

    for row in (
        cell49_payload_signal_rows
    ):

        print(
            "- cell:",
            row[
                "cell_index"
            ],
            "| terms:",
            row[
                "payload_terms"
            ],
        )

else:

    print(
        "NONE"
    )


print(
    "\n=== FORMULA RESOLUTION ==="
)

print(
    "Status:",
    cell49_golden_case_id_formula_status
)

print(
    "Resolved builder:",
    cell49_resolved_builder_name
)

print(
    "Exact builders:",
    len(
        cell49_exact_builder_matches
    )
)

print(
    "Independent reproduction failures:",
    len(
        cell49_independent_reproduction_failures
    )
)


print(
    "\n=== CODE CELL 49 GATES ==="
)

for name, passed in (
    cell_49_gates.items()
):

    print(
        name,
        ":",
        "PASS"
        if passed
        else "FAIL"
    )


print(
    "\nCODE CELL 49 GATE:",
    "PASS"
    if cell_49_pass
    else "FAIL"
)


assert cell_49_pass, (
    "Golden Case ID formula is not yet exactly resolved. "
    "Do not materialize rebuilt Exercise Golden Cases. "
    "Repair this same Code Cell 49 using the actual notebook "
    "formula evidence printed above until all 38 legacy IDs "
    "are independently reproducible."
)
```

<h3>Exercise-Unit Golden Case Schema Compatibility Audit</h3>

<p>
Cell 49 فرمول واقعی <code>golden_case_id</code> را با بازتولید 38/38 مورد حل کرد.
اما حل Formula شناسه به‌تنهایی برای ساخت Caseهای جدید کافی نیست.
</p>

<p>
Golden Caseهای قدیمی در Stratum
<code>exercise_or_self_assessment</code>
همگی Record-level بودند:
</p>

<ul>
<li><code>candidate_identity = {"document_id": ...}</code></li>
<li><code>evidence_document_count = 1</code></li>
<li><code>evidence_unit = BOOK_REGION</code></li>
</ul>

<p>
Population جدید در سطح <strong>Exercise Unit</strong> است و یک Case می‌تواند
چندین Evidence Document داشته باشد.
بنابراین قبل از Materialization باید از کد واقعی Notebook مشخص شود که:
</p>

<ul>
<li>آیا Contract فعلی Golden Case تعداد Evidence را محدود به یک Record می‌کند یا خیر؛</li>
<li>آیا <code>candidate_identity</code> الزاماً باید <code>document_id</code> داشته باشد؛</li>
<li>آیا Builderهای فعلی برای Candidate Identity فقط Record-level هستند؛</li>
<li>آیا <code>evidence_unit</code> دارای Enum یا Constraint صریح است؛</li>
<li>آیا Schema version فعلی را می‌توان بدون نقض Contract برای Exercise Unit استفاده کرد.</li>
</ul>

<p>
اگر Contract موجود Multi-document / Unit-level identity را صریحاً پشتیبانی نکند،
این Cell آن را به‌عنوان
<code>SCHEMA_VERSION_BUMP_REQUIRED</code>
علامت می‌زند.
هیچ Schema جدیدی در همین Cell تعریف نمی‌شود.
</p>

<p>
هیچ Golden Case جدید، Task، Target یا Pilot در این مرحله ساخته نمی‌شود.
</p>

```python
# Code Cell 50

# ============================================================
# Exercise-Unit Golden Case Schema Compatibility Audit
#
# Goal:
#   determine whether the EXISTING Golden Case schema /
#   builder contract can safely represent rebuilt Exercise
#   Units with multi-document evidence and unit-level identity.
#
# READ-ONLY.
#
# NO new schema definition.
# NO case materialization.
# NO supersession.
# NO Task rebuild.
# NO Target rebuild.
# NO Pilot rebuild.
# NO question generation.
# NO artifact writes.
# ============================================================


# ============================================================
# 1) Reuse actual NB12 code already loaded by Cell 49
# ============================================================

assert (
    cell49_notebook_path.exists()
)


assert (
    len(
        cell49_code_cells
    )
    > 0
)


# ============================================================
# 2) Search exact source definitions relevant to compatibility
# ============================================================

CELL50_SOURCE_TERMS = [
    "GOLDEN_CASE_RECORD_CONTRACT_V1_DRAFT",
    "GOLDEN_CASE_SCHEMA_VERSION",
    "def golden_candidate_identity",
    "def stable_candidate_identity_record",
    "def resolve_case_evidence_document_ids",
    "evidence_unit",
    "evidence_document_count",
    "evidence_document_ids",
    "candidate_identity",
]


cell50_relevant_source_cells = []


for code_cell in (
    cell49_code_cells
):

    source = (
        code_cell[
            "source"
        ]
    )


    matched_terms = [
        term
        for term
        in CELL50_SOURCE_TERMS
        if (
            term
            in
            source
        )
    ]


    if not matched_terms:
        continue


    cell50_relevant_source_cells.append(
        {
            "cell_index":
                code_cell[
                    "cell_index"
                ],

            "matched_terms":
                matched_terms,

            "source":
                source,
        }
    )


# ============================================================
# 3) Extract narrow source snippets around exact definitions
# ============================================================

CELL50_DEFINITION_ANCHORS = [
    "GOLDEN_CASE_RECORD_CONTRACT_V1_DRAFT",
    "GOLDEN_CASE_SCHEMA_VERSION",
    "def golden_candidate_identity",
    "def stable_candidate_identity_record",
    "def resolve_case_evidence_document_ids",
]


cell50_definition_snippets = []


for source_row in (
    cell50_relevant_source_cells
):

    source_lines = (
        source_row[
            "source"
        ].splitlines()
    )


    for anchor in (
        CELL50_DEFINITION_ANCHORS
    ):

        anchor_indexes = [
            index
            for index, line
            in enumerate(
                source_lines
            )
            if (
                anchor
                in
                line
            )
        ]


        for anchor_index in (
            anchor_indexes
        ):

            start_index = max(
                0,
                anchor_index
                - 5,
            )


            end_index = min(
                len(
                    source_lines
                ),
                anchor_index
                + 45,
            )


            cell50_definition_snippets.append(
                {
                    "cell_index":
                        source_row[
                            "cell_index"
                        ],

                    "anchor":
                        anchor,

                    "snippet":
                        "\n".join(
                            source_lines[
                                start_index:
                                end_index
                            ]
                        ),
                }
            )


# ============================================================
# 4) Actual runtime objects
# ============================================================

cell50_runtime_contract = globals().get(
    "GOLDEN_CASE_RECORD_CONTRACT_V1_DRAFT"
)


cell50_runtime_schema_version = globals().get(
    "GOLDEN_CASE_SCHEMA_VERSION"
)


cell50_runtime_golden_candidate_identity = globals().get(
    "golden_candidate_identity"
)


cell50_runtime_stable_candidate_identity_record = globals().get(
    "stable_candidate_identity_record"
)


cell50_runtime_resolve_case_evidence_document_ids = globals().get(
    "resolve_case_evidence_document_ids"
)


# ============================================================
# 5) Contract structural inspection
# ============================================================

cell50_contract_present = isinstance(
    cell50_runtime_contract,
    dict,
)


cell50_contract_top_level_keys = (
    sorted(
        cell50_runtime_contract.keys()
    )
    if cell50_contract_present
    else []
)


cell50_contract_required_fields = (
    list(
        cell50_runtime_contract.get(
            "required_fields",
            []
        )
    )
    if cell50_contract_present
    else []
)


# ------------------------------------------------------------
# Recursively search contract for explicit constraints.
# ------------------------------------------------------------

def cell50_find_key_value_occurrences(
    obj,
    target_key_terms,
    path="$",
):

    hits = []


    if isinstance(
        obj,
        dict,
    ):

        for key, value in (
            obj.items()
        ):

            key_text = str(
                key
            )


            if any(
                term
                in
                key_text

                for term
                in target_key_terms
            ):

                hits.append(
                    {
                        "path":
                            (
                                path
                                +
                                "["
                                +
                                repr(
                                    key
                                )
                                +
                                "]"
                            ),

                        "key":
                            key,

                        "value":
                            value,
                    }
                )


            hits.extend(
                cell50_find_key_value_occurrences(
                    value,
                    target_key_terms,
                    path=(
                        path
                        +
                        "["
                        +
                        repr(
                            key
                        )
                        +
                        "]"
                    ),
                )
            )


    elif isinstance(
        obj,
        list,
    ):

        for index, value in enumerate(
            obj
        ):

            hits.extend(
                cell50_find_key_value_occurrences(
                    value,
                    target_key_terms,
                    path=(
                        path
                        +
                        "["
                        +
                        str(
                            index
                        )
                        +
                        "]"
                    ),
                )
            )


    return hits


CELL50_CONSTRAINT_KEY_TERMS = [
    "evidence",
    "candidate",
    "identity",
    "document",
    "unit",
    "count",
    "schema",
]


cell50_contract_constraint_hits = (
    cell50_find_key_value_occurrences(
        cell50_runtime_contract,
        CELL50_CONSTRAINT_KEY_TERMS,
    )
    if cell50_contract_present
    else []
)


# ============================================================
# 6) Inspect function signatures and code metadata
# ============================================================

def cell50_callable_metadata(
    function_object,
):

    if not callable(
        function_object
    ):

        return None


    code_object = getattr(
        function_object,
        "__code__",
        None,
    )


    if (
        code_object
        is None
    ):

        return None


    return {
        "name":
            getattr(
                function_object,
                "__name__",
                None,
            ),

        "arg_names":
            list(
                code_object.co_varnames[
                    :code_object.co_argcount
                ]
            ),

        "referenced_names":
            list(
                code_object.co_names
            ),

        "constants":
            [
                value
                for value
                in code_object.co_consts
                if isinstance(
                    value,
                    (
                        str,
                        int,
                        float,
                        bool,
                        type(
                            None
                        ),
                    ),
                )
            ],
    }


cell50_candidate_identity_function_metadata = (
    cell50_callable_metadata(
        cell50_runtime_golden_candidate_identity
    )
)


cell50_stable_candidate_identity_function_metadata = (
    cell50_callable_metadata(
        cell50_runtime_stable_candidate_identity_record
    )
)


cell50_evidence_resolver_function_metadata = (
    cell50_callable_metadata(
        cell50_runtime_resolve_case_evidence_document_ids
    )
)


# ============================================================
# 7) Behavioral audit of CURRENT candidate identity function
#
# Use one selected rebuilt Exercise Unit as probe.
#
# This is read-only.
# ============================================================

cell50_probe_population_record = (
    selected_exercise_unit_sample_records[
        0
    ]
)


cell50_probe_population_unit_id = (
    cell50_probe_population_record[
        "population_unit_id"
    ]
)


cell50_probe_item = {
    "population_unit_id":
        cell50_probe_population_unit_id,

    "work_id":
        cell50_probe_population_record[
            "work_id"
        ],

    "edition_id":
        cell50_probe_population_record[
            "edition_id"
        ],

    "heading_path":
        list(
            cell50_probe_population_record[
                "heading_path"
            ]
        ),

    "evidence_document_ids":
        list(
            cell50_probe_population_record[
                "evidence_document_ids"
            ]
        ),
}


cell50_candidate_identity_probe_result = None

cell50_candidate_identity_probe_error = None


if callable(
    cell50_runtime_golden_candidate_identity
):

    try:

        cell50_candidate_identity_probe_result = (
            cell50_runtime_golden_candidate_identity(
                "book",
                "exercise_or_self_assessment",
                cell50_probe_item,
            )
        )


    except Exception as exc:

        cell50_candidate_identity_probe_error = {
            "error_type":
                type(
                    exc
                ).__name__,

            "error_message":
                str(
                    exc
                ),
        }


# ============================================================
# 8) Behavioral audit of stable_candidate_identity_record
# ============================================================

cell50_stable_candidate_probe_result = None

cell50_stable_candidate_probe_error = None


if callable(
    cell50_runtime_stable_candidate_identity_record
):

    try:

        cell50_stable_candidate_probe_result = (
            cell50_runtime_stable_candidate_identity_record(
                "book",
                "exercise_or_self_assessment",
                cell50_probe_item,
            )
        )


    except Exception as exc:

        cell50_stable_candidate_probe_error = {
            "error_type":
                type(
                    exc
                ).__name__,

            "error_message":
                str(
                    exc
                ),
        }


# ============================================================
# 9) Behavioral audit of current evidence resolver
# ============================================================

cell50_evidence_resolver_probe_result = None

cell50_evidence_resolver_probe_error = None


if callable(
    cell50_runtime_resolve_case_evidence_document_ids
):

    try:

        cell50_evidence_resolver_probe_result = (
            cell50_runtime_resolve_case_evidence_document_ids(
                "book",
                "exercise_or_self_assessment",
                cell50_probe_item,
            )
        )


    except Exception as exc:

        cell50_evidence_resolver_probe_error = {
            "error_type":
                type(
                    exc
                ).__name__,

            "error_message":
                str(
                    exc
                ),
        }


# ============================================================
# 10) Existing Exercise Case empirical constraints
# ============================================================

cell50_legacy_evidence_count_distribution = Counter(
    record[
        "evidence_document_count"
    ]
    for record
    in cell48_legacy_exercise_case_records
)


cell50_legacy_candidate_identity_key_distribution = Counter(
    tuple(
        sorted(
            record[
                "candidate_identity"
            ].keys()
        )
    )
    for record
    in cell48_legacy_exercise_case_records
)


cell50_legacy_evidence_unit_distribution = Counter(
    record[
        "evidence_unit"
    ]
    for record
    in cell48_legacy_exercise_case_records
)


# ============================================================
# 11) Rebuilt sample evidence distribution
# ============================================================

cell50_rebuilt_evidence_count_distribution = Counter(
    record[
        "evidence_document_count"
    ]
    for record
    in selected_exercise_unit_sample_records
)


cell50_rebuilt_multi_document_case_count = sum(
    1
    for record
    in selected_exercise_unit_sample_records
    if (
        record[
            "evidence_document_count"
        ]
        > 1
    )
)


cell50_rebuilt_max_evidence_document_count = max(
    record[
        "evidence_document_count"
    ]
    for record
    in selected_exercise_unit_sample_records
)


# ============================================================
# 12) Explicit contract-signal checks
#
# We distinguish:
#
# A) empirical legacy behavior
# B) explicit runtime/source support
#
# Absence of an explicit one-document restriction does NOT
# automatically prove multi-document compatibility.
# ============================================================

cell50_contract_serialized = (
    stable_json_key(
        cell50_runtime_contract
    )
    if cell50_contract_present
    else ""
)


CELL50_ONE_DOCUMENT_RESTRICTION_SIGNALS = [
    "exactly_one",
    "single_document",
    "one_document",
    "evidence_document_count\":1",
    "evidence_document_count\": 1",
]


cell50_explicit_single_document_restriction_signals = [
    signal
    for signal
    in CELL50_ONE_DOCUMENT_RESTRICTION_SIGNALS
    if (
        signal
        in
        cell50_contract_serialized
    )
]


# ============================================================
# 13) Identity compatibility classification
# ============================================================

cell50_current_candidate_identity_supports_unit_probe = (
    isinstance(
        cell50_candidate_identity_probe_result,
        dict,
    )

    and

    (
        "population_unit_id"
        in
        cell50_candidate_identity_probe_result
    )
)


cell50_current_stable_identity_supports_unit_probe = (
    isinstance(
        cell50_stable_candidate_probe_result,
        dict,
    )

    and

    isinstance(
        cell50_stable_candidate_probe_result.get(
            "candidate_identity"
        ),
        dict,
    )

    and

    (
        "population_unit_id"
        in
        cell50_stable_candidate_probe_result[
            "candidate_identity"
        ]
    )
)


cell50_current_evidence_resolver_supports_multi_document_probe = (
    isinstance(
        cell50_evidence_resolver_probe_result,
        list,
    )

    and

    cell50_evidence_resolver_probe_result
    ==
    cell50_probe_population_record[
        "evidence_document_ids"
    ]
)


# ============================================================
# 14) Overall compatibility decision
#
# Existing schema may be reused only if ALL are evidenced:
#
# - no explicit single-document restriction
# - current candidate identity supports unit identity
# - current stable identity builder supports unit identity
# - evidence resolver returns full unit evidence set
#
# Otherwise a schema/builder revision is required.
# ============================================================

cell50_no_explicit_single_document_restriction = (
    len(
        cell50_explicit_single_document_restriction_signals
    )
    == 0
)


cell50_existing_schema_directly_compatible = all(
    [
        cell50_no_explicit_single_document_restriction,
        cell50_current_candidate_identity_supports_unit_probe,
        cell50_current_stable_identity_supports_unit_probe,
        cell50_current_evidence_resolver_supports_multi_document_probe,
    ]
)


if (
    cell50_existing_schema_directly_compatible
):

    cell50_compatibility_status = (
        "CURRENT_SCHEMA_AND_BUILDERS_DIRECTLY_SUPPORT_EXERCISE_UNITS"
    )


else:

    cell50_compatibility_status = (
        "SCHEMA_OR_BUILDER_REVISION_REQUIRED_BEFORE_MATERIALIZATION"
    )


# ============================================================
# 15) Determine exact reasons requiring revision
# ============================================================

cell50_revision_reasons = []


if not (
    cell50_no_explicit_single_document_restriction
):

    cell50_revision_reasons.append(
        "EXPLICIT_SINGLE_DOCUMENT_CONSTRAINT_PRESENT"
    )


if not (
    cell50_current_candidate_identity_supports_unit_probe
):

    cell50_revision_reasons.append(
        "CURRENT_CANDIDATE_IDENTITY_DOES_NOT_SUPPORT_POPULATION_UNIT_ID"
    )


if not (
    cell50_current_stable_identity_supports_unit_probe
):

    cell50_revision_reasons.append(
        "CURRENT_STABLE_CANDIDATE_IDENTITY_BUILDER_IS_RECORD_LEVEL"
    )


if not (
    cell50_current_evidence_resolver_supports_multi_document_probe
):

    cell50_revision_reasons.append(
        "CURRENT_EVIDENCE_RESOLVER_DOES_NOT_RETURN_FULL_EXERCISE_UNIT_EVIDENCE"
    )


# ============================================================
# 16) Compatibility audit registry
# ============================================================

EXERCISE_GOLDEN_CASE_SCHEMA_COMPATIBILITY_AUDIT_V1_DRAFT = {
    "version":
        "exercise_golden_case_schema_compatibility_audit_v1_draft",

    "status":
        cell50_compatibility_status,

    "current_case_schema_version":
        cell50_runtime_schema_version,

    "legacy_exercise_case_count":
        len(
            cell48_legacy_exercise_case_records
        ),

    "rebuilt_sample_count":
        len(
            selected_exercise_unit_sample_records
        ),

    "rebuilt_multi_document_case_count":
        cell50_rebuilt_multi_document_case_count,

    "rebuilt_max_evidence_document_count":
        cell50_rebuilt_max_evidence_document_count,

    "candidate_identity_probe_supported":
        cell50_current_candidate_identity_supports_unit_probe,

    "stable_identity_probe_supported":
        cell50_current_stable_identity_supports_unit_probe,

    "evidence_resolver_multi_document_probe_supported":
        cell50_current_evidence_resolver_supports_multi_document_probe,

    "revision_reasons":
        list(
            cell50_revision_reasons
        ),

    "new_schema_definition_status":
        "NOT_STARTED",

    "case_materialization_status":
        "NOT_STARTED",

    "legacy_supersession_status":
        "NOT_STARTED",

    "task_rebuild_status":
        "NOT_STARTED",

    "target_rebuild_status":
        "NOT_STARTED",

    "pilot_rebuild_status":
        "NOT_STARTED",

    "question_generation_status":
        "NOT_STARTED",

    "artifact_write_status":
        "NONE",
}


# ============================================================
# 17) Mutation guardrails
# ============================================================

cell50_golden_case_count_before = len(
    golden_case_records
)


cell50_auto_case_count_before = len(
    auto_case_records
)


cell50_task_count_before = len(
    task_draft_plan_rows
)


cell50_target_count_before = len(
    target_evidence_records
)


cell50_sample_ids_before = set(
    selected_exercise_population_unit_ids
)


# ============================================================
# 18) Validation gates
#
# NOTE:
# This Cell is allowed to PASS with either compatibility
# outcome. Its job is to RESOLVE compatibility, not force
# compatibility.
# ============================================================

runtime_contract_present_gate = (
    cell50_contract_present
)


schema_version_present_gate = (
    isinstance(
        cell50_runtime_schema_version,
        str,
    )

    and

    bool(
        cell50_runtime_schema_version
    )
)


required_functions_present_gate = all(
    callable(
        function_object
    )

    for function_object
    in [
        cell50_runtime_golden_candidate_identity,
        cell50_runtime_stable_candidate_identity_record,
        cell50_runtime_resolve_case_evidence_document_ids,
    ]
)


actual_source_definitions_found_gate = (
    len(
        cell50_definition_snippets
    )
    > 0
)


legacy_empirical_contract_reproduced_gate = (
    cell50_legacy_evidence_count_distribution
    ==
    Counter(
        {
            1:
                38
        }
    )

    and

    cell50_legacy_candidate_identity_key_distribution
    ==
    Counter(
        {
            (
                "document_id",
            ):
                38
        }
    )
)


rebuilt_sample_is_multi_document_gate = (
    cell50_rebuilt_multi_document_case_count
    > 0

    and

    cell50_rebuilt_max_evidence_document_count
    > 1
)


compatibility_status_resolved_gate = (
    cell50_compatibility_status
    in {
        "CURRENT_SCHEMA_AND_BUILDERS_DIRECTLY_SUPPORT_EXERCISE_UNITS",
        "SCHEMA_OR_BUILDER_REVISION_REQUIRED_BEFORE_MATERIALIZATION",
    }
)


revision_reason_consistency_gate = (
    (
        cell50_existing_schema_directly_compatible

        and

        len(
            cell50_revision_reasons
        )
        == 0
    )

    or

    (
        not
        cell50_existing_schema_directly_compatible

        and

        len(
            cell50_revision_reasons
        )
        > 0
    )
)


golden_cases_not_mutated_gate = (
    len(
        golden_case_records
    )
    ==
    cell50_golden_case_count_before
)


auto_cases_not_mutated_gate = (
    len(
        auto_case_records
    )
    ==
    cell50_auto_case_count_before
)


tasks_not_mutated_gate = (
    len(
        task_draft_plan_rows
    )
    ==
    cell50_task_count_before
)


targets_not_mutated_gate = (
    len(
        target_evidence_records
    )
    ==
    cell50_target_count_before
)


sample_not_mutated_gate = (
    set(
        selected_exercise_population_unit_ids
    )
    ==
    cell50_sample_ids_before
)


new_schema_not_started_gate = (
    EXERCISE_GOLDEN_CASE_SCHEMA_COMPATIBILITY_AUDIT_V1_DRAFT[
        "new_schema_definition_status"
    ]
    ==
    "NOT_STARTED"
)


case_materialization_not_started_gate = (
    EXERCISE_GOLDEN_CASE_SCHEMA_COMPATIBILITY_AUDIT_V1_DRAFT[
        "case_materialization_status"
    ]
    ==
    "NOT_STARTED"
)


supersession_not_started_gate = (
    EXERCISE_GOLDEN_CASE_SCHEMA_COMPATIBILITY_AUDIT_V1_DRAFT[
        "legacy_supersession_status"
    ]
    ==
    "NOT_STARTED"
)


task_rebuild_not_started_gate = (
    EXERCISE_GOLDEN_CASE_SCHEMA_COMPATIBILITY_AUDIT_V1_DRAFT[
        "task_rebuild_status"
    ]
    ==
    "NOT_STARTED"
)


target_rebuild_not_started_gate = (
    EXERCISE_GOLDEN_CASE_SCHEMA_COMPATIBILITY_AUDIT_V1_DRAFT[
        "target_rebuild_status"
    ]
    ==
    "NOT_STARTED"
)


pilot_rebuild_not_started_gate = (
    EXERCISE_GOLDEN_CASE_SCHEMA_COMPATIBILITY_AUDIT_V1_DRAFT[
        "pilot_rebuild_status"
    ]
    ==
    "NOT_STARTED"
)


question_generation_not_started_gate = (
    EXERCISE_GOLDEN_CASE_SCHEMA_COMPATIBILITY_AUDIT_V1_DRAFT[
        "question_generation_status"
    ]
    ==
    "NOT_STARTED"
)


no_artifact_write_gate = (
    EXERCISE_GOLDEN_CASE_SCHEMA_COMPATIBILITY_AUDIT_V1_DRAFT[
        "artifact_write_status"
    ]
    ==
    "NONE"
)


cell_50_gates = {
    "current_golden_case_runtime_contract_present":
        runtime_contract_present_gate,

    "current_golden_case_schema_version_present":
        schema_version_present_gate,

    "candidate_identity_and_evidence_builder_functions_present":
        required_functions_present_gate,

    "actual_notebook_source_definitions_found":
        actual_source_definitions_found_gate,

    "legacy_exercise_record_level_contract_empirically_reproduced":
        legacy_empirical_contract_reproduced_gate,

    "rebuilt_exercise_sample_confirmed_multi_document":
        rebuilt_sample_is_multi_document_gate,

    "schema_compatibility_status_resolved_without_guessing":
        compatibility_status_resolved_gate,

    "revision_reasons_match_compatibility_result":
        revision_reason_consistency_gate,

    "golden_case_records_not_mutated":
        golden_cases_not_mutated_gate,

    "auto_case_records_not_mutated":
        auto_cases_not_mutated_gate,

    "task_records_not_mutated":
        tasks_not_mutated_gate,

    "target_records_not_mutated":
        targets_not_mutated_gate,

    "rebuilt_sample_not_mutated":
        sample_not_mutated_gate,

    "new_schema_definition_not_started":
        new_schema_not_started_gate,

    "new_case_materialization_not_started":
        case_materialization_not_started_gate,

    "legacy_supersession_not_started":
        supersession_not_started_gate,

    "task_rebuild_not_started":
        task_rebuild_not_started_gate,

    "target_rebuild_not_started":
        target_rebuild_not_started_gate,

    "pilot_rebuild_not_started":
        pilot_rebuild_not_started_gate,

    "question_generation_not_started":
        question_generation_not_started_gate,

    "no_artifact_write":
        no_artifact_write_gate,
}


cell_50_pass = all(
    cell_50_gates.values()
)


# ============================================================
# 19) Output
# ============================================================

print(
    "=== EXERCISE-UNIT GOLDEN CASE SCHEMA COMPATIBILITY AUDIT ==="
)


print(
    "Current schema version:",
    cell50_runtime_schema_version
)


print(
    "Legacy Exercise cases:",
    len(
        cell48_legacy_exercise_case_records
    )
)


print(
    "Legacy evidence-document-count distribution:",
    dict(
        sorted(
            cell50_legacy_evidence_count_distribution.items()
        )
    )
)


print(
    "Legacy candidate-identity key distribution:",
    dict(
        cell50_legacy_candidate_identity_key_distribution
    )
)


print(
    "Legacy evidence-unit distribution:",
    dict(
        cell50_legacy_evidence_unit_distribution
    )
)


print(
    "\n=== REBUILT SAMPLE ==="
)


print(
    "Selected Exercise Units:",
    len(
        selected_exercise_unit_sample_records
    )
)


print(
    "Multi-document units:",
    cell50_rebuilt_multi_document_case_count
)


print(
    "Maximum evidence-document count:",
    cell50_rebuilt_max_evidence_document_count
)


print(
    "Evidence-document-count distribution:",
    dict(
        sorted(
            cell50_rebuilt_evidence_count_distribution.items()
        )
    )
)


print(
    "\n=== CURRENT BUILDER PROBES ==="
)


print(
    "golden_candidate_identity result:",
    cell50_candidate_identity_probe_result
)


print(
    "golden_candidate_identity error:",
    cell50_candidate_identity_probe_error
)


print(
    "stable_candidate_identity_record result:",
    cell50_stable_candidate_probe_result
)


print(
    "stable_candidate_identity_record error:",
    cell50_stable_candidate_probe_error
)


print(
    "resolve_case_evidence_document_ids result:",
    cell50_evidence_resolver_probe_result
)


print(
    "resolve_case_evidence_document_ids error:",
    cell50_evidence_resolver_probe_error
)


print(
    "\n=== EXPLICIT CONTRACT CONSTRAINT SIGNALS ==="
)


print(
    "Single-document restriction signals:",
    cell50_explicit_single_document_restriction_signals
)


print(
    "Contract required fields:",
    cell50_contract_required_fields
)


print(
    "\n=== RELEVANT SOURCE DEFINITIONS ==="
)


for index, row in enumerate(
    cell50_definition_snippets,
    start=1,
):

    print(
        "\nSNIPPET",
        index,
        "| cell:",
        row[
            "cell_index"
        ],
        "| anchor:",
        row[
            "anchor"
        ],
    )

    print(
        row[
            "snippet"
        ]
    )


print(
    "\n=== COMPATIBILITY DECISION ==="
)


print(
    "Current candidate identity supports Exercise Unit:",
    cell50_current_candidate_identity_supports_unit_probe
)


print(
    "Current stable identity builder supports Exercise Unit:",
    cell50_current_stable_identity_supports_unit_probe
)


print(
    "Current evidence resolver returns full Exercise Unit evidence:",
    cell50_current_evidence_resolver_supports_multi_document_probe
)


print(
    "Status:",
    cell50_compatibility_status
)


print(
    "Revision reasons:",
    cell50_revision_reasons
)


print(
    "\n=== CODE CELL 50 GATES ==="
)


for name, passed in (
    cell_50_gates.items()
):

    print(
        name,
        ":",
        "PASS"
        if passed
        else "FAIL"
    )


print(
    "\nCODE CELL 50 GATE:",
    "PASS"
    if cell_50_pass
    else "FAIL"
)


assert cell_50_pass, (
    "Exercise-unit Golden Case compatibility audit failed. "
    "Do not materialize new Exercise Golden Cases until the "
    "existing schema/builders have been conclusively classified "
    "as compatible or revision-required."
)
```

```python
# Code Cell 51

# ============================================================
# Exercise-Unit Golden Case V2 Contract + Builder Freeze
#
# Scope:
#   source_type = book
#   stratum = exercise_or_self_assessment
#   identity_level = EXERCISE_UNIT
#
# IMPORTANT:
#   Existing V1 Golden Case contract and builders are NOT
#   modified.
#
# THIS CELL:
#   - defines V2 Exercise-unit identity contract
#   - defines V2 candidate identity builder
#   - defines V2 candidate hash builder
#   - defines V2 Golden Case ID builder
#   - defines V2 evidence resolver
#   - defines V2 candidate snapshot builder
#   - validates all builders over selected 38 Exercise Units
#
# NO Golden Case records materialized.
# NO legacy cases superseded.
# NO downstream registry mutation.
# NO Task rebuild.
# NO Target rebuild.
# NO Pilot rebuild.
# NO question generation.
# NO artifact writes.
# ============================================================


# ============================================================
# 1) Scoped V2 schema constants
# ============================================================

EXERCISE_UNIT_GOLDEN_CASE_SCHEMA_VERSION = (
    "golden_case_record_v2_exercise_unit_draft"
)


EXERCISE_UNIT_CANDIDATE_IDENTITY_VERSION = (
    "exercise_unit_candidate_identity_v1"
)


EXERCISE_UNIT_EVIDENCE_UNIT = (
    "BOOK_EXERCISE_UNIT"
)


EXERCISE_UNIT_SOURCE_TYPE = (
    "book"
)


EXERCISE_UNIT_STRATUM = (
    "exercise_or_self_assessment"
)


# ============================================================
# 2) V2 Case Contract
#
# This does not replace GOLDEN_CASE_RECORD_CONTRACT_V1_DRAFT.
# ============================================================

EXERCISE_UNIT_GOLDEN_CASE_RECORD_CONTRACT_V2_DRAFT = {
    "version":
        EXERCISE_UNIT_GOLDEN_CASE_SCHEMA_VERSION,

    "status":
        "BUILDER_VALIDATION_PENDING",

    "scope": {
        "source_type":
            EXERCISE_UNIT_SOURCE_TYPE,

        "stratum":
            EXERCISE_UNIT_STRATUM,

        "identity_level":
            "EXERCISE_UNIT",

        "applies_to_other_strata":
            False,
    },

    "case_identity_policy": {
        "algorithm":
            "SHA256",

        "inputs": [
            "case_schema_version",
            "source_type",
            "stratum",
            "candidate_identity_sha256",
        ],

        "digest_prefix_length":
            32,

        "id_prefix":
            "golden-case:",

        "randomness":
            False,
    },

    "candidate_identity_policy": {
        "version":
            EXERCISE_UNIT_CANDIDATE_IDENTITY_VERSION,

        "required_fields": [
            "identity_version",
            "identity_level",
            "population_unit_id",
            "work_id",
            "edition_id",
            "heading_path",
            "evidence_document_ids",
        ],

        "evidence_order_significant":
            True,

        "mutable_sampling_status_excluded":
            True,

        "mutable_golden_membership_excluded":
            True,
    },

    "evidence_policy": {
        "evidence_unit":
            EXERCISE_UNIT_EVIDENCE_UNIT,

        "minimum_document_count":
            1,

        "multi_document_supported":
            True,

        "cross_unit_document_overlap":
            "FORBIDDEN",

        "accepted_corpus_only":
            True,

        "locator_required":
            True,

        "lineage_required":
            True,

        "rights_propagation":
            "SOURCE_ONLY_NO_INFERENCE",

        "attribution_propagation":
            "SOURCE_ONLY_NO_INFERENCE",
    },

    "selection_policy": {
        "sampling_policy_source":
            EXERCISE_UNIT_REBUILT_SAMPLE_V1_DRAFT[
                "version"
            ],

        "population_snapshot_sha256":
            exercise_population_snapshot_sha256,

        "sample_snapshot_sha256":
            exercise_sample_snapshot_sha256,

        "population_count":
            45,

        "sampling_target":
            38,

        "selection_method":
            (
                "LARGEST_REMAINDER_WORK_PROPORTIONAL_"
                "DETERMINISTIC_COVERAGE_SAFE_EXCLUSION"
            ),

        "selection_rank_semantics":
            (
                "DETERMINISTIC_SERIALIZATION_ORDER_ONLY_"
                "NOT_QUALITY_PRIORITY"
            ),

        "random_sampling":
            False,

        "random_seed":
            None,
    },

    "required_case_fields": [
        "golden_case_id",
        "case_schema_version",
        "source_type",
        "stratum",
        "selection_rank",
        "candidate_identity",
        "candidate_identity_sha256",
        "candidate_snapshot",
        "membership_status",
        "final_golden_membership",
        "evidence_unit",
        "evidence_document_ids",
        "evidence_document_count",
        "evidence_references",
        "review_status",
        "question_generation_status",
        "expected_answer_generation_status",
        "selection_provenance",
    ],

    "evidence_reference_required_fields":
        list(
            GOLDEN_CASE_RECORD_CONTRACT_V1_DRAFT[
                "evidence_reference_required_fields"
            ]
        ),

    "materialization_status":
        "NOT_STARTED",

    "legacy_supersession_status":
        "NOT_STARTED",

    "task_rebuild_status":
        "NOT_STARTED",

    "target_rebuild_status":
        "NOT_STARTED",

    "pilot_rebuild_status":
        "NOT_STARTED",

    "question_generation_status":
        "NOT_STARTED",

    "artifact_write_status":
        "NONE",
}


# ============================================================
# 3) V2 candidate identity builder
# ============================================================

def build_exercise_unit_candidate_identity_v2(
    item,
):

    required_input_fields = [
        "population_unit_id",
        "work_id",
        "edition_id",
        "heading_path",
        "evidence_document_ids",
    ]


    missing_fields = [
        field
        for field
        in required_input_fields
        if (
            field
            not in item
        )
    ]


    if (
        missing_fields
    ):

        raise AssertionError(
            "Exercise-unit candidate identity input is missing: "
            f"{missing_fields}"
        )


    evidence_document_ids = list(
        item[
            "evidence_document_ids"
        ]
    )


    if not (
        evidence_document_ids
    ):

        raise AssertionError(
            "Exercise-unit candidate identity cannot have "
            "empty evidence_document_ids."
        )


    if (
        len(
            evidence_document_ids
        )
        !=
        len(
            set(
                evidence_document_ids
            )
        )
    ):

        raise AssertionError(
            "Duplicate document IDs inside Exercise-unit identity."
        )


    return {
        "identity_version":
            EXERCISE_UNIT_CANDIDATE_IDENTITY_VERSION,

        "identity_level":
            "EXERCISE_UNIT",

        "population_unit_id":
            item[
                "population_unit_id"
            ],

        "work_id":
            item[
                "work_id"
            ],

        "edition_id":
            item[
                "edition_id"
            ],

        "heading_path":
            list(
                item[
                    "heading_path"
                ]
            ),

        "evidence_document_ids":
            evidence_document_ids,
    }


# ============================================================
# 4) Stable V2 candidate identity hash
# ============================================================

def stable_exercise_unit_candidate_identity_record_v2(
    item,
):

    identity = (
        build_exercise_unit_candidate_identity_v2(
            item
        )
    )


    identity_json = stable_json_key(
        identity
    )


    identity_sha256 = hashlib.sha256(
        identity_json.encode(
            "utf-8"
        )
    ).hexdigest()


    return {
        "identity":
            identity,

        "identity_json":
            identity_json,

        "identity_sha256":
            identity_sha256,
    }


# ============================================================
# 5) V2 Golden Case ID builder
#
# Same proven algorithm as build_golden_case_id,
# but explicitly scoped to the new V2 schema version.
#
# Existing build_golden_case_id is NOT modified.
# ============================================================

def build_exercise_unit_golden_case_id_v2(
    candidate_identity_sha256,
):

    case_identity_payload = {
        "case_schema_version":
            EXERCISE_UNIT_GOLDEN_CASE_SCHEMA_VERSION,

        "source_type":
            EXERCISE_UNIT_SOURCE_TYPE,

        "stratum":
            EXERCISE_UNIT_STRATUM,

        "candidate_identity_sha256":
            candidate_identity_sha256,
    }


    canonical = stable_json_key(
        case_identity_payload
    )


    digest = hashlib.sha256(
        canonical.encode(
            "utf-8"
        )
    ).hexdigest()


    return (
        "golden-case:"
        +
        digest[
            :32
        ]
    )


# ============================================================
# 6) V2 evidence resolver
# ============================================================

def resolve_exercise_unit_evidence_document_ids_v2(
    item,
):

    values = (
        item.get(
            "evidence_document_ids"
        )
        or []
    )


    if not isinstance(
        values,
        list,
    ):

        raise AssertionError(
            "Exercise-unit evidence_document_ids must be a list."
        )


    resolved_ids = list(
        values
    )


    if not (
        resolved_ids
    ):

        raise AssertionError(
            "Exercise Unit must resolve at least one evidence document."
        )


    if (
        len(
            resolved_ids
        )
        !=
        len(
            set(
                resolved_ids
            )
        )
    ):

        raise AssertionError(
            "Exercise Unit contains duplicate evidence document IDs."
        )


    return resolved_ids


# ============================================================
# 7) V2 candidate snapshot builder
#
# Snapshot is descriptive and auditable.
# It is NOT the primary stable identity.
# ============================================================

def build_exercise_unit_candidate_snapshot_v2(
    item,
):

    evidence_document_ids = (
        resolve_exercise_unit_evidence_document_ids_v2(
            item
        )
    )


    return {
        "identity_level":
            "EXERCISE_UNIT",

        "population_unit_id":
            item[
                "population_unit_id"
            ],

        "work_id":
            item[
                "work_id"
            ],

        "edition_id":
            item[
                "edition_id"
            ],

        "heading_path":
            list(
                item[
                    "heading_path"
                ]
            ),

        "terminal_markers":
            list(
                item[
                    "terminal_markers"
                ]
            ),

        "boundary_source":
            item[
                "boundary_source"
            ],

        "evidence_document_ids":
            evidence_document_ids,

        "evidence_document_count":
            len(
                evidence_document_ids
            ),

        "exercise_core_token_count":
            item[
                "exercise_core_token_count"
            ],

        "exercise_core_evidence_sha256":
            item[
                "exercise_core_evidence_sha256"
            ],

        "population_snapshot_sha256":
            exercise_population_snapshot_sha256,

        "sample_snapshot_sha256":
            exercise_sample_snapshot_sha256,
    }


# ============================================================
# 8) V2 selection provenance builder
# ============================================================

def build_exercise_unit_selection_provenance_v2():

    return {
        "sampling_policy_version":
            EXERCISE_UNIT_REBUILT_SAMPLE_V1_DRAFT[
                "version"
            ],

        "selection_method":
            EXERCISE_UNIT_GOLDEN_CASE_RECORD_CONTRACT_V2_DRAFT[
                "selection_policy"
            ][
                "selection_method"
            ],

        "sampling_target":
            38,

        "population_count":
            45,

        "population_snapshot_sha256":
            exercise_population_snapshot_sha256,

        "sample_snapshot_sha256":
            exercise_sample_snapshot_sha256,

        "random_sampling":
            False,

        "random_seed":
            None,
    }


# ============================================================
# 9) Deterministic selection serialization ranks
#
# Ranking has NO quality semantics.
# ============================================================

cell51_ranked_selected_unit_ids = sorted(
    selected_exercise_population_unit_ids
)


cell51_selection_rank_by_population_unit_id = {
    population_unit_id:
        rank

    for rank, population_unit_id
    in enumerate(
        cell51_ranked_selected_unit_ids,
        start=1,
    )
}


# ============================================================
# 10) Validate builders over all 38 selected Exercise Units
# ============================================================

cell51_builder_validation_rows = []


cell51_validation_failures = []


for sample_record in (
    selected_exercise_unit_sample_records
):

    population_unit_id = (
        sample_record[
            "population_unit_id"
        ]
    )


    try:

        identity_record = (
            stable_exercise_unit_candidate_identity_record_v2(
                sample_record
            )
        )


        evidence_document_ids = (
            resolve_exercise_unit_evidence_document_ids_v2(
                sample_record
            )
        )


        candidate_snapshot = (
            build_exercise_unit_candidate_snapshot_v2(
                sample_record
            )
        )


        golden_case_id = (
            build_exercise_unit_golden_case_id_v2(
                identity_record[
                    "identity_sha256"
                ]
            )
        )


        selection_rank = (
            cell51_selection_rank_by_population_unit_id[
                population_unit_id
            ]
        )


    except Exception as exc:

        cell51_validation_failures.append(
            {
                "population_unit_id":
                    population_unit_id,

                "reason":
                    "BUILDER_EXCEPTION",

                "error_type":
                    type(
                        exc
                    ).__name__,

                "error_message":
                    str(
                        exc
                    ),
            }
        )

        continue


    # --------------------------------------------------------
    # Accepted corpus / evidence integrity
    # --------------------------------------------------------

    missing_corpus_ids = [
        document_id

        for document_id
        in evidence_document_ids

        if (
            document_id
            not in
            corpus_target_index
        )
    ]


    evidence_records = [
        corpus_target_index[
            document_id
        ]

        for document_id
        in evidence_document_ids

        if (
            document_id
            in
            corpus_target_index
        )
    ]


    source_type_failures = [
        record[
            "document_id"
        ]

        for record
        in evidence_records

        if (
            record[
                "source_type"
            ]
            !=
            "book"
        )
    ]


    locator_failures = [
        record[
            "document_id"
        ]

        for record
        in evidence_records

        if not (
            record[
                "locator"
            ]
        )
    ]


    lineage_failures = [
        record[
            "document_id"
        ]

        for record
        in evidence_records

        if (
            record[
                "lineage"
            ]
            is None
        )
    ]


    rights_failures = [
        record[
            "document_id"
        ]

        for record
        in evidence_records

        if (
            record[
                "rights"
            ]
            is None
        )
    ]


    attribution_failures = [
        record[
            "document_id"
        ]

        for record
        in evidence_records

        if (
            record[
                "attribution"
            ]
            is None
        )
    ]


    locator_work_ids = {
        record[
            "locator"
        ].get(
            "work_id"
        )

        for record
        in evidence_records
    }


    locator_edition_ids = {
        record[
            "locator"
        ].get(
            "edition_id"
        )

        for record
        in evidence_records
    }


    work_identity_matches = (
        locator_work_ids
        ==
        {
            sample_record[
                "work_id"
            ]
        }
    )


    edition_identity_matches = (
        locator_edition_ids
        ==
        {
            sample_record[
                "edition_id"
            ]
        }
    )


    identity_evidence_matches = (
        identity_record[
            "identity"
        ][
            "evidence_document_ids"
        ]
        ==
        evidence_document_ids
    )


    snapshot_evidence_matches = (
        candidate_snapshot[
            "evidence_document_ids"
        ]
        ==
        evidence_document_ids
    )


    evidence_count_matches = (
        candidate_snapshot[
            "evidence_document_count"
        ]
        ==
        len(
            evidence_document_ids
        )
        ==
        sample_record[
            "evidence_document_count"
        ]
    )


    cell51_builder_validation_rows.append(
        {
            "population_unit_id":
                population_unit_id,

            "selection_rank":
                selection_rank,

            "candidate_identity":
                identity_record[
                    "identity"
                ],

            "candidate_identity_sha256":
                identity_record[
                    "identity_sha256"
                ],

            "golden_case_id":
                golden_case_id,

            "candidate_snapshot":
                candidate_snapshot,

            "evidence_document_ids":
                evidence_document_ids,

            "evidence_document_count":
                len(
                    evidence_document_ids
                ),

            "missing_corpus_ids":
                missing_corpus_ids,

            "source_type_failures":
                source_type_failures,

            "locator_failures":
                locator_failures,

            "lineage_failures":
                lineage_failures,

            "rights_failures":
                rights_failures,

            "attribution_failures":
                attribution_failures,

            "work_identity_matches":
                work_identity_matches,

            "edition_identity_matches":
                edition_identity_matches,

            "identity_evidence_matches":
                identity_evidence_matches,

            "snapshot_evidence_matches":
                snapshot_evidence_matches,

            "evidence_count_matches":
                evidence_count_matches,
        }
    )


# ============================================================
# 11) Aggregate IDs / hashes
# ============================================================

cell51_candidate_identity_hashes = [
    row[
        "candidate_identity_sha256"
    ]

    for row
    in cell51_builder_validation_rows
]


cell51_proposed_v2_golden_case_ids = [
    row[
        "golden_case_id"
    ]

    for row
    in cell51_builder_validation_rows
]


cell51_existing_golden_case_ids = set(
    golden_case_records_by_id.keys()
)


cell51_v2_case_id_collisions_with_existing = sorted(
    set(
        cell51_proposed_v2_golden_case_ids
    )
    &
    cell51_existing_golden_case_ids
)


# ============================================================
# 12) Independent identity/hash/ID recomputation
# ============================================================

cell51_identity_hash_recomputation_failures = []

cell51_case_id_recomputation_failures = []


for row in (
    cell51_builder_validation_rows
):

    recomputed_identity_hash = hashlib.sha256(
        stable_json_key(
            row[
                "candidate_identity"
            ]
        ).encode(
            "utf-8"
        )
    ).hexdigest()


    if (
        recomputed_identity_hash
        !=
        row[
            "candidate_identity_sha256"
        ]
    ):

        cell51_identity_hash_recomputation_failures.append(
            row[
                "population_unit_id"
            ]
        )


    recomputed_case_id = (
        build_exercise_unit_golden_case_id_v2(
            row[
                "candidate_identity_sha256"
            ]
        )
    )


    if (
        recomputed_case_id
        !=
        row[
            "golden_case_id"
        ]
    ):

        cell51_case_id_recomputation_failures.append(
            row[
                "population_unit_id"
            ]
        )


# ============================================================
# 13) Evidence / structural failure aggregation
# ============================================================

cell51_evidence_integrity_failure_rows = []


for row in (
    cell51_builder_validation_rows
):

    failure_reasons = []


    if (
        row[
            "missing_corpus_ids"
        ]
    ):

        failure_reasons.append(
            "MISSING_CORPUS_EVIDENCE"
        )


    if (
        row[
            "source_type_failures"
        ]
    ):

        failure_reasons.append(
            "NON_BOOK_EVIDENCE"
        )


    if (
        row[
            "locator_failures"
        ]
    ):

        failure_reasons.append(
            "MISSING_LOCATOR"
        )


    if (
        row[
            "lineage_failures"
        ]
    ):

        failure_reasons.append(
            "MISSING_LINEAGE"
        )


    if (
        row[
            "rights_failures"
        ]
    ):

        failure_reasons.append(
            "MISSING_RIGHTS"
        )


    if (
        row[
            "attribution_failures"
        ]
    ):

        failure_reasons.append(
            "MISSING_ATTRIBUTION"
        )


    if not (
        row[
            "work_identity_matches"
        ]
    ):

        failure_reasons.append(
            "WORK_IDENTITY_MISMATCH"
        )


    if not (
        row[
            "edition_identity_matches"
        ]
    ):

        failure_reasons.append(
            "EDITION_IDENTITY_MISMATCH"
        )


    if not (
        row[
            "identity_evidence_matches"
        ]
    ):

        failure_reasons.append(
            "IDENTITY_EVIDENCE_MISMATCH"
        )


    if not (
        row[
            "snapshot_evidence_matches"
        ]
    ):

        failure_reasons.append(
            "SNAPSHOT_EVIDENCE_MISMATCH"
        )


    if not (
        row[
            "evidence_count_matches"
        ]
    ):

        failure_reasons.append(
            "EVIDENCE_COUNT_MISMATCH"
        )


    if (
        failure_reasons
    ):

        cell51_evidence_integrity_failure_rows.append(
            {
                "population_unit_id":
                    row[
                        "population_unit_id"
                    ],

                "failure_reasons":
                    failure_reasons,
            }
        )


# ============================================================
# 14) Rank integrity
# ============================================================

cell51_selection_ranks = sorted(
    row[
        "selection_rank"
    ]
    for row
    in cell51_builder_validation_rows
)


cell51_expected_selection_ranks = list(
    range(
        1,
        39,
    )
)


# ============================================================
# 15) V1 preservation snapshots
# ============================================================

cell51_v1_schema_version_snapshot = (
    GOLDEN_CASE_SCHEMA_VERSION
)


cell51_v1_contract_version_snapshot = (
    GOLDEN_CASE_RECORD_CONTRACT_V1_DRAFT[
        "version"
    ]
)


cell51_v1_builder_snapshot = (
    build_golden_case_id
)


cell51_golden_case_count_snapshot = len(
    golden_case_records
)


cell51_auto_case_count_snapshot = len(
    auto_case_records
)


cell51_task_count_snapshot = len(
    task_draft_plan_rows
)


cell51_target_count_snapshot = len(
    target_evidence_records
)


# ============================================================
# 16) Builder freeze registry
# ============================================================

EXERCISE_UNIT_GOLDEN_CASE_V2_BUILDER_FREEZE_DRAFT = {
    "version":
        "exercise_unit_golden_case_v2_builder_freeze_draft",

    "status":
        "VALIDATED_BUILDERS_NOT_MATERIALIZED",

    "case_schema_version":
        EXERCISE_UNIT_GOLDEN_CASE_SCHEMA_VERSION,

    "candidate_identity_version":
        EXERCISE_UNIT_CANDIDATE_IDENTITY_VERSION,

    "evidence_unit":
        EXERCISE_UNIT_EVIDENCE_UNIT,

    "selected_sample_count":
        len(
            cell51_builder_validation_rows
        ),

    "candidate_identity_hash_count":
        len(
            set(
                cell51_candidate_identity_hashes
            )
        ),

    "proposed_case_id_count":
        len(
            set(
                cell51_proposed_v2_golden_case_ids
            )
        ),

    "existing_case_id_collision_count":
        len(
            cell51_v2_case_id_collisions_with_existing
        ),

    "selection_rank_policy":
        (
            "SORTED_POPULATION_UNIT_ID_"
            "SERIALIZATION_ORDER_ONLY"
        ),

    "materialization_status":
        "NOT_STARTED",

    "legacy_supersession_status":
        "NOT_STARTED",

    "task_rebuild_status":
        "NOT_STARTED",

    "target_rebuild_status":
        "NOT_STARTED",

    "pilot_rebuild_status":
        "NOT_STARTED",

    "question_generation_status":
        "NOT_STARTED",

    "artifact_write_status":
        "NONE",
}


# ============================================================
# 17) Validation gates
# ============================================================

exact_38_builder_rows_gate = (
    len(
        cell51_builder_validation_rows
    )
    == 38

    and

    len(
        cell51_validation_failures
    )
    == 0
)


all_candidate_hashes_unique_gate = (
    len(
        set(
            cell51_candidate_identity_hashes
        )
    )
    == 38
)


all_proposed_case_ids_unique_gate = (
    len(
        set(
            cell51_proposed_v2_golden_case_ids
        )
    )
    == 38
)


no_existing_case_id_collision_gate = (
    len(
        cell51_v2_case_id_collisions_with_existing
    )
    == 0
)


identity_hashes_recomputable_gate = (
    len(
        cell51_identity_hash_recomputation_failures
    )
    == 0
)


case_ids_recomputable_gate = (
    len(
        cell51_case_id_recomputation_failures
    )
    == 0
)


all_evidence_integrity_pass_gate = (
    len(
        cell51_evidence_integrity_failure_rows
    )
    == 0
)


selection_ranks_exact_1_to_38_gate = (
    cell51_selection_ranks
    ==
    cell51_expected_selection_ranks
)


v1_schema_unchanged_gate = (
    GOLDEN_CASE_SCHEMA_VERSION
    ==
    cell51_v1_schema_version_snapshot

    and

    GOLDEN_CASE_RECORD_CONTRACT_V1_DRAFT[
        "version"
    ]
    ==
    cell51_v1_contract_version_snapshot

    and

    build_golden_case_id
    is
    cell51_v1_builder_snapshot
)


golden_case_records_not_materialized_gate = (
    len(
        golden_case_records
    )
    ==
    cell51_golden_case_count_snapshot
    ==
    724
)


auto_case_records_not_mutated_gate = (
    len(
        auto_case_records
    )
    ==
    cell51_auto_case_count_snapshot
    ==
    572
)


task_records_not_mutated_gate = (
    len(
        task_draft_plan_rows
    )
    ==
    cell51_task_count_snapshot
)


target_records_not_mutated_gate = (
    len(
        target_evidence_records
    )
    ==
    cell51_target_count_snapshot
)


materialization_not_started_gate = (
    EXERCISE_UNIT_GOLDEN_CASE_V2_BUILDER_FREEZE_DRAFT[
        "materialization_status"
    ]
    ==
    "NOT_STARTED"
)


supersession_not_started_gate = (
    EXERCISE_UNIT_GOLDEN_CASE_V2_BUILDER_FREEZE_DRAFT[
        "legacy_supersession_status"
    ]
    ==
    "NOT_STARTED"
)


task_rebuild_not_started_gate = (
    EXERCISE_UNIT_GOLDEN_CASE_V2_BUILDER_FREEZE_DRAFT[
        "task_rebuild_status"
    ]
    ==
    "NOT_STARTED"
)


target_rebuild_not_started_gate = (
    EXERCISE_UNIT_GOLDEN_CASE_V2_BUILDER_FREEZE_DRAFT[
        "target_rebuild_status"
    ]
    ==
    "NOT_STARTED"
)


pilot_rebuild_not_started_gate = (
    EXERCISE_UNIT_GOLDEN_CASE_V2_BUILDER_FREEZE_DRAFT[
        "pilot_rebuild_status"
    ]
    ==
    "NOT_STARTED"
)


question_generation_not_started_gate = (
    EXERCISE_UNIT_GOLDEN_CASE_V2_BUILDER_FREEZE_DRAFT[
        "question_generation_status"
    ]
    ==
    "NOT_STARTED"
)


no_artifact_write_gate = (
    EXERCISE_UNIT_GOLDEN_CASE_V2_BUILDER_FREEZE_DRAFT[
        "artifact_write_status"
    ]
    ==
    "NONE"
)


cell_51_gates = {
    "exact_38_selected_exercise_units_validate_under_v2_builders":
        exact_38_builder_rows_gate,

    "all_38_v2_candidate_identity_hashes_unique":
        all_candidate_hashes_unique_gate,

    "all_38_proposed_v2_golden_case_ids_unique":
        all_proposed_case_ids_unique_gate,

    "no_v2_case_id_collides_with_existing_724_cases":
        no_existing_case_id_collision_gate,

    "all_v2_candidate_identity_hashes_independently_recomputable":
        identity_hashes_recomputable_gate,

    "all_v2_golden_case_ids_independently_recomputable":
        case_ids_recomputable_gate,

    "all_38_exercise_unit_evidence_sets_preserve_corpus_and_lineage_integrity":
        all_evidence_integrity_pass_gate,

    "selection_serialization_ranks_are_exactly_1_through_38":
        selection_ranks_exact_1_to_38_gate,

    "existing_v1_schema_contract_and_builder_unchanged":
        v1_schema_unchanged_gate,

    "golden_case_records_still_724_and_not_materialized":
        golden_case_records_not_materialized_gate,

    "auto_case_records_still_572_and_not_mutated":
        auto_case_records_not_mutated_gate,

    "task_records_not_mutated":
        task_records_not_mutated_gate,

    "target_records_not_mutated":
        target_records_not_mutated_gate,

    "v2_case_materialization_not_started":
        materialization_not_started_gate,

    "legacy_supersession_not_started":
        supersession_not_started_gate,

    "task_rebuild_not_started":
        task_rebuild_not_started_gate,

    "target_rebuild_not_started":
        target_rebuild_not_started_gate,

    "pilot_rebuild_not_started":
        pilot_rebuild_not_started_gate,

    "question_generation_not_started":
        question_generation_not_started_gate,

    "no_artifact_write":
        no_artifact_write_gate,
}


cell_51_pass = all(
    cell_51_gates.values()
)


# ============================================================
# 18) Finalize in-memory builder-contract status
#
# Contract definition itself is allowed in this Cell.
# Case materialization is not.
# ============================================================

if (
    cell_51_pass
):

    EXERCISE_UNIT_GOLDEN_CASE_RECORD_CONTRACT_V2_DRAFT[
        "status"
    ] = (
        "BUILDERS_VALIDATED_NOT_MATERIALIZED"
    )


# ============================================================
# 19) Output
# ============================================================

print(
    "=== EXERCISE-UNIT GOLDEN CASE V2 BUILDER FREEZE ==="
)


print(
    "V2 schema version:",
    EXERCISE_UNIT_GOLDEN_CASE_SCHEMA_VERSION
)


print(
    "Identity version:",
    EXERCISE_UNIT_CANDIDATE_IDENTITY_VERSION
)


print(
    "Evidence unit:",
    EXERCISE_UNIT_EVIDENCE_UNIT
)


print(
    "Selected units validated:",
    len(
        cell51_builder_validation_rows
    )
)


print(
    "Builder exceptions:",
    len(
        cell51_validation_failures
    )
)


print(
    "Unique candidate identity hashes:",
    len(
        set(
            cell51_candidate_identity_hashes
        )
    )
)


print(
    "Unique proposed Golden Case IDs:",
    len(
        set(
            cell51_proposed_v2_golden_case_ids
        )
    )
)


print(
    "Collisions with current 724 Golden Case IDs:",
    len(
        cell51_v2_case_id_collisions_with_existing
    )
)


print(
    "\n=== EVIDENCE DISTRIBUTION UNDER V2 ==="
)


print(
    dict(
        sorted(
            Counter(
                row[
                    "evidence_document_count"
                ]

                for row
                in cell51_builder_validation_rows
            ).items()
        )
    )
)


print(
    "Maximum evidence documents:",
    max(
        row[
            "evidence_document_count"
        ]

        for row
        in cell51_builder_validation_rows
    )
)


print(
    "\n=== INTEGRITY ==="
)


print(
    "Evidence integrity failures:",
    len(
        cell51_evidence_integrity_failure_rows
    )
)


print(
    "Candidate identity hash recomputation failures:",
    len(
        cell51_identity_hash_recomputation_failures
    )
)


print(
    "Golden Case ID recomputation failures:",
    len(
        cell51_case_id_recomputation_failures
    )
)


print(
    "\n=== V1 PRESERVATION ==="
)


print(
    "Existing schema:",
    GOLDEN_CASE_SCHEMA_VERSION
)


print(
    "Existing case count:",
    len(
        golden_case_records
    )
)


print(
    "Existing auto case count:",
    len(
        auto_case_records
    )
)


print(
    "\n=== SAMPLE PROPOSED V2 IDENTITIES ==="
)


for row in (
    cell51_builder_validation_rows[
        :5
    ]
):

    print(
        "\nPopulation unit:",
        row[
            "population_unit_id"
        ]
    )

    print(
        "Selection rank:",
        row[
            "selection_rank"
        ]
    )

    print(
        "Candidate identity:",
        row[
            "candidate_identity"
        ]
    )

    print(
        "Candidate identity SHA256:",
        row[
            "candidate_identity_sha256"
        ]
    )

    print(
        "Proposed Golden Case ID:",
        row[
            "golden_case_id"
        ]
    )

    print(
        "Evidence documents:",
        row[
            "evidence_document_count"
        ]
    )


print(
    "\n=== CONTRACT STATUS ==="
)


print(
    EXERCISE_UNIT_GOLDEN_CASE_RECORD_CONTRACT_V2_DRAFT[
        "status"
    ]
)


print(
    "\n=== CODE CELL 51 GATES ==="
)


for name, passed in (
    cell_51_gates.items()
):

    print(
        name,
        ":",
        "PASS"
        if passed
        else "FAIL"
    )


print(
    "\nCODE CELL 51 GATE:",
    "PASS"
    if cell_51_pass
    else "FAIL"
)


assert cell_51_pass, (
    "Exercise-unit Golden Case V2 builder freeze failed. "
    "Do not materialize rebuilt Golden Cases until all 38 "
    "selected units have unique/recomputable V2 identities, "
    "full evidence integrity, zero collisions with the existing "
    "724 cases, and the V1 contract remains unchanged."
)
```

```python
# Code Cell 52 — REPAIRED AGAIN

# ============================================================
# Stage 38 Exercise-Unit Golden Case V2 Records
#
# ROOT-CAUSE REPAIR:
#
# Previous Cell 52 used corpus_target_index to build full
# Evidence References.
#
# Actual evidence showed corpus_target_index is a reduced
# Target-Selection view and does NOT contain:
#   - text_sha256
#   - raw_content_hash
#   - content_origin
#
# The existing notebook evidence resolver already depends on
# corpus_evidence_index. This repair therefore:
#
#   1) validates corpus_evidence_index as the canonical
#      Evidence Reference source;
#   2) verifies all 480 selected documents exist there;
#   3) verifies all required historical Evidence Reference
#      fields are present;
#   4) cross-checks shared fields against corpus_target_index;
#   5) stages 38 V2 Golden Cases;
#   6) independently re-validates IDs, evidence and ordering;
#   7) proves active registries remain unchanged.
#
# STAGING ONLY.
#
# NO active Golden Case mutation.
# NO supersession.
# NO Task rebuild.
# NO Target rebuild.
# NO Pilot rebuild.
# NO question generation.
# NO artifact writes.
# ============================================================


# ============================================================
# 1) Snapshot active registries BEFORE staging
# ============================================================

cell52_active_golden_count_before = len(
    golden_case_records
)

cell52_active_golden_id_count_before = len(
    golden_case_records_by_id
)

cell52_active_auto_count_before = len(
    auto_case_records
)

cell52_active_task_count_before = len(
    task_draft_plan_rows
)

cell52_active_target_count_before = len(
    target_evidence_records
)

cell52_active_pilot_count_before = len(
    pilot_selection_rows
)

cell52_selected_sample_ids_before = set(
    selected_exercise_population_unit_ids
)

cell52_active_golden_ids_before = set(
    golden_case_records_by_id.keys()
)


cell52_active_golden_snapshot_before = hashlib.sha256(
    stable_json_key(
        golden_case_records
    ).encode(
        "utf-8"
    )
).hexdigest()


cell52_active_auto_snapshot_before = hashlib.sha256(
    stable_json_key(
        auto_case_records
    ).encode(
        "utf-8"
    )
).hexdigest()


cell52_active_task_snapshot_before = hashlib.sha256(
    stable_json_key(
        task_draft_plan_rows
    ).encode(
        "utf-8"
    )
).hexdigest()


cell52_active_target_snapshot_before = hashlib.sha256(
    stable_json_key(
        target_evidence_records
    ).encode(
        "utf-8"
    )
).hexdigest()


cell52_active_pilot_snapshot_before = hashlib.sha256(
    stable_json_key(
        pilot_selection_rows
    ).encode(
        "utf-8"
    )
).hexdigest()


# ============================================================
# 2) Reproduce the actual historical Evidence Reference schema
# ============================================================

cell52_legacy_exercise_evidence_references = [
    evidence_reference

    for case_record
    in cell48_legacy_exercise_case_records

    for evidence_reference
    in case_record[
        "evidence_references"
    ]
]


cell52_legacy_evidence_reference_key_sets = [
    set(
        evidence_reference.keys()
    )

    for evidence_reference
    in cell52_legacy_exercise_evidence_references
]


assert (
    len(
        cell52_legacy_evidence_reference_key_sets
    )
    > 0
)


cell52_legacy_evidence_reference_key_union = set().union(
    *cell52_legacy_evidence_reference_key_sets
)


cell52_legacy_evidence_reference_key_intersection = set(
    cell52_legacy_evidence_reference_key_sets[
        0
    ]
)


for key_set in (
    cell52_legacy_evidence_reference_key_sets[
        1:
    ]
):

    cell52_legacy_evidence_reference_key_intersection &= (
        key_set
    )


cell52_legacy_evidence_reference_schema_uniform = (
    cell52_legacy_evidence_reference_key_union
    ==
    cell52_legacy_evidence_reference_key_intersection
)


cell52_historical_evidence_reference_fields = sorted(
    cell52_legacy_evidence_reference_key_intersection
)


# Expected from actual legacy records, not invented.
cell52_historical_evidence_reference_expected_fields = {
    "attribution",
    "content_origin",
    "document_id",
    "lineage",
    "locator",
    "raw_content_hash",
    "rights",
    "source_id",
    "source_type",
    "text_sha256",
    "title",
}


# ============================================================
# 3) Validate the real full Evidence Index
#
# corpus_evidence_index is grounded by the existing notebook's
# resolve_case_evidence_document_ids implementation.
# ============================================================

cell52_canonical_evidence_index = globals().get(
    "corpus_evidence_index"
)


cell52_canonical_evidence_index_present = (
    isinstance(
        cell52_canonical_evidence_index,
        dict,
    )
)


assert (
    cell52_canonical_evidence_index_present
), (
    "corpus_evidence_index is not available as a dict. "
    "Do not fall back to corpus_target_index for full Evidence "
    "References."
)


# ============================================================
# 4) Selected evidence universe
# ============================================================

cell52_selected_unique_evidence_document_ids = sorted(
    {
        document_id

        for sample_record
        in selected_exercise_unit_sample_records

        for document_id
        in sample_record[
            "evidence_document_ids"
        ]
    }
)


# ============================================================
# 5) Canonical Evidence Index preflight
# ============================================================

cell52_missing_from_canonical_evidence_index = []

cell52_missing_historical_fields_in_canonical_index = []

cell52_canonical_evidence_document_identity_failures = []


for document_id in (
    cell52_selected_unique_evidence_document_ids
):

    if (
        document_id
        not in
        cell52_canonical_evidence_index
    ):

        cell52_missing_from_canonical_evidence_index.append(
            document_id
        )

        continue


    evidence_record = (
        cell52_canonical_evidence_index[
            document_id
        ]
    )


    if not isinstance(
        evidence_record,
        dict,
    ):

        cell52_missing_historical_fields_in_canonical_index.append(
            {
                "document_id":
                    document_id,

                "reason":
                    "EVIDENCE_RECORD_NOT_DICT",
            }
        )

        continue


    missing_fields = sorted(
        cell52_historical_evidence_reference_expected_fields
        -
        set(
            evidence_record.keys()
        )
    )


    if (
        missing_fields
    ):

        cell52_missing_historical_fields_in_canonical_index.append(
            {
                "document_id":
                    document_id,

                "missing_fields":
                    missing_fields,

                "available_fields":
                    sorted(
                        evidence_record.keys()
                    ),
            }
        )


    if (
        evidence_record.get(
            "document_id"
        )
        !=
        document_id
    ):

        cell52_canonical_evidence_document_identity_failures.append(
            {
                "lookup_document_id":
                    document_id,

                "record_document_id":
                    evidence_record.get(
                        "document_id"
                    ),
            }
        )


# ============================================================
# 6) Cross-check canonical Evidence Index against the lighter
# Target Index on every field both views actually share.
#
# This proves we are switching representation, not switching
# underlying documents.
# ============================================================

cell52_shared_target_evidence_fields = {
    "document_id",
    "source_id",
    "source_type",
    "title",
    "locator",
    "lineage",
    "attribution",
    "rights",
}


cell52_target_vs_evidence_index_mismatches = []

cell52_missing_selected_ids_from_target_index = []


for document_id in (
    cell52_selected_unique_evidence_document_ids
):

    if (
        document_id
        not in
        corpus_target_index
    ):

        cell52_missing_selected_ids_from_target_index.append(
            document_id
        )

        continue


    if (
        document_id
        not in
        cell52_canonical_evidence_index
    ):

        continue


    target_record = (
        corpus_target_index[
            document_id
        ]
    )


    evidence_record = (
        cell52_canonical_evidence_index[
            document_id
        ]
    )


    field_mismatches = []


    for field_name in sorted(
        cell52_shared_target_evidence_fields
    ):

        if (
            field_name
            not in
            target_record
        ):

            field_mismatches.append(
                {
                    "field":
                        field_name,

                    "reason":
                        "TARGET_VIEW_FIELD_MISSING",
                }
            )

            continue


        if (
            field_name
            not in
            evidence_record
        ):

            field_mismatches.append(
                {
                    "field":
                        field_name,

                    "reason":
                        "EVIDENCE_VIEW_FIELD_MISSING",
                }
            )

            continue


        if (
            stable_json_key(
                target_record[
                    field_name
                ]
            )
            !=
            stable_json_key(
                evidence_record[
                    field_name
                ]
            )
        ):

            field_mismatches.append(
                {
                    "field":
                        field_name,

                    "reason":
                        "VALUE_MISMATCH",
                }
            )


    if (
        field_mismatches
    ):

        cell52_target_vs_evidence_index_mismatches.append(
            {
                "document_id":
                    document_id,

                "field_mismatches":
                    field_mismatches,
            }
        )


# ============================================================
# 7) Safe recursive metadata clone
# ============================================================

def cell52_safe_clone(
    value,
):

    if isinstance(
        value,
        dict,
    ):

        return {
            key:
                cell52_safe_clone(
                    nested_value
                )

            for key, nested_value
            in value.items()
        }


    if isinstance(
        value,
        list,
    ):

        return [
            cell52_safe_clone(
                nested_value
            )

            for nested_value
            in value
        ]


    if isinstance(
        value,
        tuple,
    ):

        return tuple(
            cell52_safe_clone(
                nested_value
            )

            for nested_value
            in value
        )


    if isinstance(
        value,
        set,
    ):

        return set(
            cell52_safe_clone(
                nested_value
            )

            for nested_value
            in value
        )


    return value


# ============================================================
# 8) Canonical Evidence Reference builder
#
# Uses ONLY corpus_evidence_index.
# ============================================================

def build_exercise_v2_evidence_reference(
    document_id,
):

    if (
        document_id
        not in
        cell52_canonical_evidence_index
    ):

        raise AssertionError(
            "Evidence document missing from corpus_evidence_index: "
            f"{document_id}"
        )


    source_record = (
        cell52_canonical_evidence_index[
            document_id
        ]
    )


    missing_fields = sorted(
        cell52_historical_evidence_reference_expected_fields
        -
        set(
            source_record.keys()
        )
    )


    if (
        missing_fields
    ):

        raise AssertionError(
            "Canonical evidence record is missing required "
            "historical Evidence Reference fields: "
            f"document_id={document_id}; "
            f"missing={missing_fields}"
        )


    if (
        source_record[
            "document_id"
        ]
        !=
        document_id
    ):

        raise AssertionError(
            "Canonical evidence document identity mismatch: "
            f"{document_id}"
        )


    return {
        field_name:
            cell52_safe_clone(
                source_record[
                    field_name
                ]
            )

        for field_name
        in sorted(
            cell52_historical_evidence_reference_expected_fields
        )
    }


# ============================================================
# 9) Evidence-reference builder preflight for all 480 docs
# ============================================================

cell52_evidence_reference_preflight_failures = []


for document_id in (
    cell52_selected_unique_evidence_document_ids
):

    try:

        built_reference = (
            build_exercise_v2_evidence_reference(
                document_id
            )
        )


    except Exception as exc:

        cell52_evidence_reference_preflight_failures.append(
            {
                "document_id":
                    document_id,

                "error_type":
                    type(
                        exc
                    ).__name__,

                "error_message":
                    str(
                        exc
                    ),
            }
        )

        continue


    if (
        set(
            built_reference.keys()
        )
        !=
        cell52_historical_evidence_reference_expected_fields
    ):

        cell52_evidence_reference_preflight_failures.append(
            {
                "document_id":
                    document_id,

                "error_type":
                    "REFERENCE_SCHEMA_MISMATCH",

                "error_message":
                    str(
                        sorted(
                            built_reference.keys()
                        )
                    ),
            }
        )


# ============================================================
# 10) Build one staged V2 Golden Case
# ============================================================

def build_staged_exercise_v2_golden_case(
    sample_record,
):

    population_unit_id = (
        sample_record[
            "population_unit_id"
        ]
    )


    identity_record = (
        stable_exercise_unit_candidate_identity_record_v2(
            sample_record
        )
    )


    evidence_document_ids = (
        resolve_exercise_unit_evidence_document_ids_v2(
            sample_record
        )
    )


    evidence_references = [
        build_exercise_v2_evidence_reference(
            document_id
        )

        for document_id
        in evidence_document_ids
    ]


    candidate_snapshot = (
        build_exercise_unit_candidate_snapshot_v2(
            sample_record
        )
    )


    golden_case_id = (
        build_exercise_unit_golden_case_id_v2(
            identity_record[
                "identity_sha256"
            ]
        )
    )


    selection_rank = (
        cell51_selection_rank_by_population_unit_id[
            population_unit_id
        ]
    )


    return {
        "golden_case_id":
            golden_case_id,

        "case_schema_version":
            EXERCISE_UNIT_GOLDEN_CASE_SCHEMA_VERSION,

        "source_type":
            EXERCISE_UNIT_SOURCE_TYPE,

        "stratum":
            EXERCISE_UNIT_STRATUM,

        "selection_rank":
            selection_rank,

        "candidate_identity":
            cell52_safe_clone(
                identity_record[
                    "identity"
                ]
            ),

        "candidate_identity_sha256":
            identity_record[
                "identity_sha256"
            ],

        "candidate_snapshot":
            cell52_safe_clone(
                candidate_snapshot
            ),

        "membership_status":
            "AUTO_GOLDEN_SELECTED",

        "final_golden_membership":
            True,

        "evidence_unit":
            EXERCISE_UNIT_EVIDENCE_UNIT,

        "evidence_document_ids":
            list(
                evidence_document_ids
            ),

        "evidence_document_count":
            len(
                evidence_document_ids
            ),

        "evidence_references":
            evidence_references,

        "review_status":
            "NOT_REQUIRED",

        "question_generation_status":
            "NOT_STARTED",

        "expected_answer_generation_status":
            "NOT_STARTED",

        "selection_provenance":
            cell52_safe_clone(
                build_exercise_unit_selection_provenance_v2()
            ),
    }


# ============================================================
# 11) Materialize 38 cases into isolated staging
# ============================================================

staged_exercise_v2_golden_case_records = []

staged_exercise_v2_golden_case_records_by_id = {}

cell52_staging_build_failures = []


for sample_record in (
    selected_exercise_unit_sample_records
):

    try:

        staged_case = (
            build_staged_exercise_v2_golden_case(
                sample_record
            )
        )


    except Exception as exc:

        cell52_staging_build_failures.append(
            {
                "population_unit_id":
                    sample_record[
                        "population_unit_id"
                    ],

                "heading_path":
                    sample_record[
                        "heading_path"
                    ],

                "evidence_document_count":
                    sample_record[
                        "evidence_document_count"
                    ],

                "error_type":
                    type(
                        exc
                    ).__name__,

                "error_message":
                    str(
                        exc
                    ),
            }
        )

        continue


    staged_case_id = (
        staged_case[
            "golden_case_id"
        ]
    )


    if (
        staged_case_id
        in
        staged_exercise_v2_golden_case_records_by_id
    ):

        cell52_staging_build_failures.append(
            {
                "population_unit_id":
                    sample_record[
                        "population_unit_id"
                    ],

                "error_type":
                    "DUPLICATE_STAGED_CASE_ID",

                "error_message":
                    staged_case_id,
            }
        )

        continue


    staged_exercise_v2_golden_case_records.append(
        staged_case
    )


    staged_exercise_v2_golden_case_records_by_id[
        staged_case_id
    ] = staged_case


staged_exercise_v2_golden_case_records = sorted(
    staged_exercise_v2_golden_case_records,
    key=lambda record:
        record[
            "selection_rank"
        ],
)


# ============================================================
# 12) Required-field validation
# ============================================================

cell52_required_case_fields = set(
    EXERCISE_UNIT_GOLDEN_CASE_RECORD_CONTRACT_V2_DRAFT[
        "required_case_fields"
    ]
)


cell52_required_evidence_fields = set(
    EXERCISE_UNIT_GOLDEN_CASE_RECORD_CONTRACT_V2_DRAFT[
        "evidence_reference_required_fields"
    ]
)


cell52_missing_case_field_rows = []

cell52_missing_evidence_field_rows = []


for case_record in (
    staged_exercise_v2_golden_case_records
):

    missing_case_fields = sorted(
        cell52_required_case_fields
        -
        set(
            case_record.keys()
        )
    )


    if (
        missing_case_fields
    ):

        cell52_missing_case_field_rows.append(
            {
                "golden_case_id":
                    case_record[
                        "golden_case_id"
                    ],

                "missing_fields":
                    missing_case_fields,
            }
        )


    for evidence_reference in (
        case_record[
            "evidence_references"
        ]
    ):

        missing_evidence_fields = sorted(
            cell52_required_evidence_fields
            -
            set(
                evidence_reference.keys()
            )
        )


        if (
            missing_evidence_fields
        ):

            cell52_missing_evidence_field_rows.append(
                {
                    "golden_case_id":
                        case_record[
                            "golden_case_id"
                        ],

                    "document_id":
                        evidence_reference.get(
                            "document_id"
                        ),

                    "missing_fields":
                        missing_evidence_fields,
                }
            )


# ============================================================
# 13) Independent Candidate Identity hash recomputation
# ============================================================

cell52_candidate_identity_recompute_failures = []


for case_record in (
    staged_exercise_v2_golden_case_records
):

    independent_candidate_hash = hashlib.sha256(
        stable_json_key(
            case_record[
                "candidate_identity"
            ]
        ).encode(
            "utf-8"
        )
    ).hexdigest()


    if (
        independent_candidate_hash
        !=
        case_record[
            "candidate_identity_sha256"
        ]
    ):

        cell52_candidate_identity_recompute_failures.append(
            case_record[
                "golden_case_id"
            ]
        )


# ============================================================
# 14) Independent Golden Case ID recomputation
#
# Does NOT call the V2 case-ID builder.
# ============================================================

cell52_case_id_recompute_failures = []


for case_record in (
    staged_exercise_v2_golden_case_records
):

    independent_case_identity_payload = {
        "case_schema_version":
            case_record[
                "case_schema_version"
            ],

        "source_type":
            case_record[
                "source_type"
            ],

        "stratum":
            case_record[
                "stratum"
            ],

        "candidate_identity_sha256":
            case_record[
                "candidate_identity_sha256"
            ],
    }


    independent_case_digest = hashlib.sha256(
        stable_json_key(
            independent_case_identity_payload
        ).encode(
            "utf-8"
        )
    ).hexdigest()


    independent_case_id = (
        "golden-case:"
        +
        independent_case_digest[
            :32
        ]
    )


    if (
        independent_case_id
        !=
        case_record[
            "golden_case_id"
        ]
    ):

        cell52_case_id_recompute_failures.append(
            case_record[
                "golden_case_id"
            ]
        )


# ============================================================
# 15) Independent Evidence Reference exactness check
#
# Directly compare each staged reference to canonical
# corpus_evidence_index values.
# ============================================================

cell52_evidence_reference_mismatch_rows = []


for case_record in (
    staged_exercise_v2_golden_case_records
):

    for evidence_reference in (
        case_record[
            "evidence_references"
        ]
    ):

        document_id = (
            evidence_reference[
                "document_id"
            ]
        )


        if (
            document_id
            not in
            cell52_canonical_evidence_index
        ):

            cell52_evidence_reference_mismatch_rows.append(
                {
                    "golden_case_id":
                        case_record[
                            "golden_case_id"
                    ],

                    "document_id":
                        document_id,

                    "reason":
                        "MISSING_CANONICAL_EVIDENCE_RECORD",
                }
            )

            continue


        canonical_record = (
            cell52_canonical_evidence_index[
                document_id
            ]
        )


        field_mismatches = []


        for field_name in sorted(
            cell52_historical_evidence_reference_expected_fields
        ):

            if (
                stable_json_key(
                    evidence_reference[
                        field_name
                    ]
                )
                !=
                stable_json_key(
                    canonical_record[
                        field_name
                    ]
                )
            ):

                field_mismatches.append(
                    field_name
                )


        if (
            field_mismatches
        ):

            cell52_evidence_reference_mismatch_rows.append(
                {
                    "golden_case_id":
                        case_record[
                            "golden_case_id"
                    ],

                    "document_id":
                        document_id,

                    "field_mismatches":
                        field_mismatches,
                }
            )


# ============================================================
# 16) Evidence membership / order / count integrity
# ============================================================

cell52_evidence_integrity_failure_rows = []


for case_record in (
    staged_exercise_v2_golden_case_records
):

    candidate_identity = (
        case_record[
            "candidate_identity"
        ]
    )


    candidate_snapshot = (
        case_record[
            "candidate_snapshot"
        ]
    )


    evidence_document_ids = list(
        case_record[
            "evidence_document_ids"
        ]
    )


    evidence_reference_ids = [
        reference[
            "document_id"
        ]

        for reference
        in case_record[
            "evidence_references"
        ]
    ]


    failure_reasons = []


    if (
        candidate_identity[
            "evidence_document_ids"
        ]
        !=
        evidence_document_ids
    ):

        failure_reasons.append(
            "IDENTITY_EVIDENCE_ORDER_MISMATCH"
        )


    if (
        candidate_snapshot[
            "evidence_document_ids"
        ]
        !=
        evidence_document_ids
    ):

        failure_reasons.append(
            "SNAPSHOT_EVIDENCE_ORDER_MISMATCH"
        )


    if (
        evidence_reference_ids
        !=
        evidence_document_ids
    ):

        failure_reasons.append(
            "REFERENCE_EVIDENCE_ORDER_MISMATCH"
        )


    if not (
        case_record[
            "evidence_document_count"
        ]
        ==
        len(
            evidence_document_ids
        )
        ==
        len(
            case_record[
                "evidence_references"
            ]
        )
    ):

        failure_reasons.append(
            "EVIDENCE_COUNT_MISMATCH"
        )


    if (
        len(
            evidence_document_ids
        )
        !=
        len(
            set(
                evidence_document_ids
            )
        )
    ):

        failure_reasons.append(
            "DUPLICATE_EVIDENCE_DOCUMENT_ID"
        )


    if (
        failure_reasons
    ):

        cell52_evidence_integrity_failure_rows.append(
            {
                "golden_case_id":
                    case_record[
                        "golden_case_id"
                    ],

                "failure_reasons":
                    failure_reasons,
            }
        )


# ============================================================
# 17) Exact sample linkage
# ============================================================

cell52_population_unit_ids_from_staged_cases = {
    case_record[
        "candidate_identity"
    ][
        "population_unit_id"
    ]

    for case_record
    in staged_exercise_v2_golden_case_records
}


cell52_missing_selected_population_units = sorted(
    set(
        selected_exercise_population_unit_ids
    )
    -
    cell52_population_unit_ids_from_staged_cases
)


cell52_extra_staged_population_units = sorted(
    cell52_population_unit_ids_from_staged_cases
    -
    set(
        selected_exercise_population_unit_ids
    )
)


# ============================================================
# 18) Selection-rank integrity
# ============================================================

cell52_selection_ranks = [
    case_record[
        "selection_rank"
    ]

    for case_record
    in staged_exercise_v2_golden_case_records
]


cell52_rank_population_pairs = [
    (
        case_record[
            "selection_rank"
        ],

        case_record[
            "candidate_identity"
        ][
            "population_unit_id"
        ],
    )

    for case_record
    in staged_exercise_v2_golden_case_records
]


cell52_expected_rank_population_pairs = [
    (
        rank,
        population_unit_id,
    )

    for rank, population_unit_id
    in enumerate(
        sorted(
            selected_exercise_population_unit_ids
        ),
        start=1,
    )
]


# ============================================================
# 19) Collision audit
# ============================================================

cell52_staged_case_ids = {
    record[
        "golden_case_id"
    ]

    for record
    in staged_exercise_v2_golden_case_records
}


cell52_case_id_collisions_with_active = sorted(
    cell52_staged_case_ids
    &
    cell52_active_golden_ids_before
)


# ============================================================
# 20) Status audit
# ============================================================

cell52_membership_status_counts = Counter(
    record[
        "membership_status"
    ]

    for record
    in staged_exercise_v2_golden_case_records
)


cell52_final_membership_counts = Counter(
    record[
        "final_golden_membership"
    ]

    for record
    in staged_exercise_v2_golden_case_records
)


cell52_review_status_counts = Counter(
    record[
        "review_status"
    ]

    for record
    in staged_exercise_v2_golden_case_records
)


cell52_question_status_counts = Counter(
    record[
        "question_generation_status"
    ]

    for record
    in staged_exercise_v2_golden_case_records
)


cell52_expected_answer_status_counts = Counter(
    record[
        "expected_answer_generation_status"
    ]

    for record
    in staged_exercise_v2_golden_case_records
)


# ============================================================
# 21) Staged snapshot SHA256
# ============================================================

cell52_staged_snapshot_payload = [
    {
        "golden_case_id":
            record[
                "golden_case_id"
            ],

        "candidate_identity_sha256":
            record[
                "candidate_identity_sha256"
            ],

        "selection_rank":
            record[
                "selection_rank"
            ],

        "population_unit_id":
            record[
                "candidate_identity"
            ][
                "population_unit_id"
            ],

        "evidence_document_ids":
            record[
                "evidence_document_ids"
            ],
    }

    for record
    in staged_exercise_v2_golden_case_records
]


cell52_staged_snapshot_sha256 = hashlib.sha256(
    stable_json_key(
        cell52_staged_snapshot_payload
    ).encode(
        "utf-8"
    )
).hexdigest()


# ============================================================
# 22) Recompute active-registry content hashes
# ============================================================

cell52_active_golden_snapshot_after = hashlib.sha256(
    stable_json_key(
        golden_case_records
    ).encode(
        "utf-8"
    )
).hexdigest()


cell52_active_auto_snapshot_after = hashlib.sha256(
    stable_json_key(
        auto_case_records
    ).encode(
        "utf-8"
    )
).hexdigest()


cell52_active_task_snapshot_after = hashlib.sha256(
    stable_json_key(
        task_draft_plan_rows
    ).encode(
        "utf-8"
    )
).hexdigest()


cell52_active_target_snapshot_after = hashlib.sha256(
    stable_json_key(
        target_evidence_records
    ).encode(
        "utf-8"
    )
).hexdigest()


cell52_active_pilot_snapshot_after = hashlib.sha256(
    stable_json_key(
        pilot_selection_rows
    ).encode(
        "utf-8"
    )
).hexdigest()


cell52_active_registry_unchanged = (
    len(
        golden_case_records
    )
    ==
    cell52_active_golden_count_before
    ==
    724

    and

    len(
        golden_case_records_by_id
    )
    ==
    cell52_active_golden_id_count_before
    ==
    724

    and

    len(
        auto_case_records
    )
    ==
    cell52_active_auto_count_before
    ==
    572

    and

    len(
        task_draft_plan_rows
    )
    ==
    cell52_active_task_count_before
    ==
    724

    and

    len(
        target_evidence_records
    )
    ==
    cell52_active_target_count_before
    ==
    572

    and

    len(
        pilot_selection_rows
    )
    ==
    cell52_active_pilot_count_before
    ==
    23

    and

    set(
        selected_exercise_population_unit_ids
    )
    ==
    cell52_selected_sample_ids_before

    and

    set(
        golden_case_records_by_id.keys()
    )
    ==
    cell52_active_golden_ids_before

    and

    cell52_active_golden_snapshot_after
    ==
    cell52_active_golden_snapshot_before

    and

    cell52_active_auto_snapshot_after
    ==
    cell52_active_auto_snapshot_before

    and

    cell52_active_task_snapshot_after
    ==
    cell52_active_task_snapshot_before

    and

    cell52_active_target_snapshot_after
    ==
    cell52_active_target_snapshot_before

    and

    cell52_active_pilot_snapshot_after
    ==
    cell52_active_pilot_snapshot_before
)


# ============================================================
# 23) Staging registry
# ============================================================

EXERCISE_UNIT_GOLDEN_CASE_V2_STAGING_V1_DRAFT = {
    "version":
        "exercise_unit_golden_case_v2_staging_v1_draft",

    "status":
        (
            "STAGED_AND_VALIDATED_NOT_ACTIVATED"
            if (
                len(
                    staged_exercise_v2_golden_case_records
                )
                == 38

                and

                len(
                    cell52_staging_build_failures
                )
                == 0
            )
            else
            "STAGING_FAILED_NOT_ACTIVATED"
        ),

    "case_schema_version":
        EXERCISE_UNIT_GOLDEN_CASE_SCHEMA_VERSION,

    "canonical_evidence_source":
        "corpus_evidence_index",

    "population_snapshot_sha256":
        exercise_population_snapshot_sha256,

    "sample_snapshot_sha256":
        exercise_sample_snapshot_sha256,

    "staged_case_snapshot_sha256":
        cell52_staged_snapshot_sha256,

    "selected_unique_evidence_document_count":
        len(
            cell52_selected_unique_evidence_document_ids
        ),

    "staged_case_count":
        len(
            staged_exercise_v2_golden_case_records
        ),

    "build_failure_count":
        len(
            cell52_staging_build_failures
        ),

    "active_registry_mutation":
        False,

    "legacy_supersession_status":
        "NOT_STARTED",

    "task_rebuild_status":
        "NOT_STARTED",

    "target_rebuild_status":
        "NOT_STARTED",

    "pilot_rebuild_status":
        "NOT_STARTED",

    "question_generation_status":
        "NOT_STARTED",

    "artifact_write_status":
        "NONE",
}


# ============================================================
# 24) Validation gates
# ============================================================

legacy_reference_schema_gate = (
    cell52_legacy_evidence_reference_schema_uniform

    and

    set(
        cell52_historical_evidence_reference_fields
    )
    ==
    cell52_historical_evidence_reference_expected_fields
)


canonical_evidence_index_gate = (
    cell52_canonical_evidence_index_present
)


exact_480_selected_evidence_documents_gate = (
    len(
        cell52_selected_unique_evidence_document_ids
    )
    == 480
)


all_480_in_canonical_evidence_index_gate = (
    len(
        cell52_missing_from_canonical_evidence_index
    )
    == 0
)


all_480_have_full_reference_fields_gate = (
    len(
        cell52_missing_historical_fields_in_canonical_index
    )
    == 0
)


all_480_canonical_ids_match_gate = (
    len(
        cell52_canonical_evidence_document_identity_failures
    )
    == 0
)


target_and_evidence_views_agree_gate = (
    len(
        cell52_missing_selected_ids_from_target_index
    )
    == 0

    and

    len(
        cell52_target_vs_evidence_index_mismatches
    )
    == 0
)


reference_builder_preflight_gate = (
    len(
        cell52_evidence_reference_preflight_failures
    )
    == 0
)


exact_38_staged_gate = (
    len(
        staged_exercise_v2_golden_case_records
    )
    == 38

    and

    len(
        staged_exercise_v2_golden_case_records_by_id
    )
    == 38

    and

    len(
        cell52_staging_build_failures
    )
    == 0
)


all_staged_ids_unique_gate = (
    len(
        cell52_staged_case_ids
    )
    == 38
)


all_required_case_fields_gate = (
    len(
        cell52_missing_case_field_rows
    )
    == 0
)


all_required_evidence_fields_gate = (
    len(
        cell52_missing_evidence_field_rows
    )
    == 0
)


identity_hash_recompute_gate = (
    len(
        cell52_candidate_identity_recompute_failures
    )
    == 0
)


case_id_recompute_gate = (
    len(
        cell52_case_id_recompute_failures
    )
    == 0
)


evidence_reference_exactness_gate = (
    len(
        cell52_evidence_reference_mismatch_rows
    )
    == 0
)


evidence_integrity_gate = (
    len(
        cell52_evidence_integrity_failure_rows
    )
    == 0
)


exact_sample_linkage_gate = (
    len(
        cell52_missing_selected_population_units
    )
    == 0

    and

    len(
        cell52_extra_staged_population_units
    )
    == 0

    and

    len(
        cell52_population_unit_ids_from_staged_cases
    )
    == 38
)


selection_rank_gate = (
    cell52_selection_ranks
    ==
    list(
        range(
            1,
            39,
        )
    )

    and

    cell52_rank_population_pairs
    ==
    cell52_expected_rank_population_pairs
)


no_active_case_id_collision_gate = (
    len(
        cell52_case_id_collisions_with_active
    )
    == 0
)


status_contract_gate = (
    cell52_membership_status_counts
    ==
    Counter(
        {
            "AUTO_GOLDEN_SELECTED":
                38
        }
    )

    and

    cell52_final_membership_counts
    ==
    Counter(
        {
            True:
                38
        }
    )

    and

    cell52_review_status_counts
    ==
    Counter(
        {
            "NOT_REQUIRED":
                38
        }
    )

    and

    cell52_question_status_counts
    ==
    Counter(
        {
            "NOT_STARTED":
                38
        }
    )

    and

    cell52_expected_answer_status_counts
    ==
    Counter(
        {
            "NOT_STARTED":
                38
        }
    )
)


active_registries_unchanged_gate = (
    cell52_active_registry_unchanged
)


legacy_supersession_not_started_gate = (
    EXERCISE_UNIT_GOLDEN_CASE_V2_STAGING_V1_DRAFT[
        "legacy_supersession_status"
    ]
    ==
    "NOT_STARTED"
)


task_rebuild_not_started_gate = (
    EXERCISE_UNIT_GOLDEN_CASE_V2_STAGING_V1_DRAFT[
        "task_rebuild_status"
    ]
    ==
    "NOT_STARTED"
)


target_rebuild_not_started_gate = (
    EXERCISE_UNIT_GOLDEN_CASE_V2_STAGING_V1_DRAFT[
        "target_rebuild_status"
    ]
    ==
    "NOT_STARTED"
)


pilot_rebuild_not_started_gate = (
    EXERCISE_UNIT_GOLDEN_CASE_V2_STAGING_V1_DRAFT[
        "pilot_rebuild_status"
    ]
    ==
    "NOT_STARTED"
)


question_generation_not_started_gate = (
    EXERCISE_UNIT_GOLDEN_CASE_V2_STAGING_V1_DRAFT[
        "question_generation_status"
    ]
    ==
    "NOT_STARTED"
)


no_artifact_write_gate = (
    EXERCISE_UNIT_GOLDEN_CASE_V2_STAGING_V1_DRAFT[
        "artifact_write_status"
    ]
    ==
    "NONE"
)


cell_52_gates = {
    "legacy_exercise_evidence_reference_schema_exactly_reproduced":
        legacy_reference_schema_gate,

    "canonical_corpus_evidence_index_present":
        canonical_evidence_index_gate,

    "exact_480_unique_selected_evidence_documents":
        exact_480_selected_evidence_documents_gate,

    "all_480_selected_documents_exist_in_canonical_evidence_index":
        all_480_in_canonical_evidence_index_gate,

    "all_480_canonical_records_have_full_evidence_reference_fields":
        all_480_have_full_reference_fields_gate,

    "all_480_canonical_document_ids_match_lookup_identity":
        all_480_canonical_ids_match_gate,

    "target_view_and_canonical_evidence_view_agree_on_shared_fields":
        target_and_evidence_views_agree_gate,

    "all_480_full_evidence_references_pass_builder_preflight":
        reference_builder_preflight_gate,

    "exact_38_v2_golden_cases_staged":
        exact_38_staged_gate,

    "all_38_staged_golden_case_ids_unique":
        all_staged_ids_unique_gate,

    "all_v2_required_case_fields_present":
        all_required_case_fields_gate,

    "all_v2_required_evidence_reference_fields_present":
        all_required_evidence_fields_gate,

    "all_candidate_identity_hashes_independently_recompute":
        identity_hash_recompute_gate,

    "all_v2_golden_case_ids_independently_recompute":
        case_id_recompute_gate,

    "all_staged_evidence_references_exactly_match_canonical_evidence_index":
        evidence_reference_exactness_gate,

    "all_staged_case_evidence_order_and_counts_are_consistent":
        evidence_integrity_gate,

    "staged_cases_exactly_cover_selected_38_population_units":
        exact_sample_linkage_gate,

    "selection_ranks_exactly_match_deterministic_1_to_38_order":
        selection_rank_gate,

    "no_staged_v2_case_id_collides_with_active_724":
        no_active_case_id_collision_gate,

    "all_staged_membership_and_generation_statuses_match_contract":
        status_contract_gate,

    "all_active_case_task_target_pilot_registries_content_unchanged":
        active_registries_unchanged_gate,

    "legacy_supersession_not_started":
        legacy_supersession_not_started_gate,

    "task_rebuild_not_started":
        task_rebuild_not_started_gate,

    "target_rebuild_not_started":
        target_rebuild_not_started_gate,

    "pilot_rebuild_not_started":
        pilot_rebuild_not_started_gate,

    "question_generation_not_started":
        question_generation_not_started_gate,

    "no_artifact_write":
        no_artifact_write_gate,
}


cell_52_pass = all(
    cell_52_gates.values()
)


# ============================================================
# 25) Output
# ============================================================

print(
    "=== EXERCISE-UNIT GOLDEN CASE V2 STAGING ==="
)


print(
    "Canonical Evidence source:",
    "corpus_evidence_index"
)


print(
    "Historical Evidence Reference fields:",
    cell52_historical_evidence_reference_fields
)


print(
    "\n=== CANONICAL EVIDENCE PREFLIGHT ==="
)


print(
    "Unique selected evidence documents:",
    len(
        cell52_selected_unique_evidence_document_ids
    )
)


print(
    "Missing from corpus_evidence_index:",
    len(
        cell52_missing_from_canonical_evidence_index
    )
)


print(
    "Records missing full Evidence Reference fields:",
    len(
        cell52_missing_historical_fields_in_canonical_index
    )
)


print(
    "Canonical document-ID failures:",
    len(
        cell52_canonical_evidence_document_identity_failures
    )
)


print(
    "Target-view / Evidence-view mismatches:",
    len(
        cell52_target_vs_evidence_index_mismatches
    )
)


print(
    "Evidence Reference builder preflight failures:",
    len(
        cell52_evidence_reference_preflight_failures
    )
)


if (
    cell52_missing_historical_fields_in_canonical_index
):

    print(
        "\nFirst canonical schema failures:"
    )

    for row in (
        cell52_missing_historical_fields_in_canonical_index[
            :10
        ]
    ):

        print(
            "-",
            row
        )


if (
    cell52_target_vs_evidence_index_mismatches
):

    print(
        "\nFirst target/evidence-view mismatches:"
    )

    for row in (
        cell52_target_vs_evidence_index_mismatches[
            :10
        ]
    ):

        print(
            "-",
            row
        )


print(
    "\n=== STAGING BUILD ==="
)


print(
    "Staged cases:",
    len(
        staged_exercise_v2_golden_case_records
    )
)


print(
    "Staging build failures:",
    len(
        cell52_staging_build_failures
    )
)


if (
    cell52_staging_build_failures
):

    print(
        "\nBUILD FAILURE DETAILS:"
    )

    for failure in (
        cell52_staging_build_failures[
            :10
        ]
    ):

        print(
            "-",
            failure
        )


print(
    "Staged snapshot SHA256:",
    cell52_staged_snapshot_sha256
)


print(
    "Unique staged Case IDs:",
    len(
        cell52_staged_case_ids
    )
)


print(
    "Collisions with active 724:",
    len(
        cell52_case_id_collisions_with_active
    )
)


print(
    "\n=== STAGED EVIDENCE ==="
)


print(
    "Evidence-document-count distribution:",
    dict(
        sorted(
            Counter(
                record[
                    "evidence_document_count"
                ]

                for record
                in staged_exercise_v2_golden_case_records
            ).items()
        )
    )
)


print(
    "Total Evidence References:",
    sum(
        record[
            "evidence_document_count"
        ]

        for record
        in staged_exercise_v2_golden_case_records
    )
)


print(
    "\n=== INDEPENDENT VALIDATION ==="
)


print(
    "Missing Case fields:",
    len(
        cell52_missing_case_field_rows
    )
)


print(
    "Missing Evidence fields:",
    len(
        cell52_missing_evidence_field_rows
    )
)


print(
    "Candidate hash failures:",
    len(
        cell52_candidate_identity_recompute_failures
    )
)


print(
    "Case-ID failures:",
    len(
        cell52_case_id_recompute_failures
    )
)


print(
    "Evidence Reference mismatches:",
    len(
        cell52_evidence_reference_mismatch_rows
    )
)


print(
    "Evidence integrity failures:",
    len(
        cell52_evidence_integrity_failure_rows
    )
)


print(
    "\n=== SAMPLE LINKAGE ==="
)


print(
    "Selected units:",
    len(
        selected_exercise_population_unit_ids
    )
)


print(
    "Staged represented units:",
    len(
        cell52_population_unit_ids_from_staged_cases
    )
)


print(
    "Missing units:",
    len(
        cell52_missing_selected_population_units
    )
)


print(
    "Extra units:",
    len(
        cell52_extra_staged_population_units
    )
)


print(
    "\n=== STATUS COUNTS ==="
)


print(
    "Membership:",
    dict(
        cell52_membership_status_counts
    )
)


print(
    "Final membership:",
    dict(
        cell52_final_membership_counts
    )
)


print(
    "Review:",
    dict(
        cell52_review_status_counts
    )
)


print(
    "Question:",
    dict(
        cell52_question_status_counts
    )
)


print(
    "Expected answer:",
    dict(
        cell52_expected_answer_status_counts
    )
)


print(
    "\n=== ACTIVE REGISTRY PRESERVATION ==="
)


print(
    "Golden cases:",
    len(
        golden_case_records
    )
)


print(
    "Auto cases:",
    len(
        auto_case_records
    )
)


print(
    "Task plans:",
    len(
        task_draft_plan_rows
    )
)


print(
    "Targets:",
    len(
        target_evidence_records
    )
)


print(
    "Pilots:",
    len(
        pilot_selection_rows
    )
)


print(
    "Active registry content unchanged:",
    cell52_active_registry_unchanged
)


print(
    "\n=== FIRST 3 STAGED CASES ==="
)


for record in (
    staged_exercise_v2_golden_case_records[
        :3
    ]
):

    print(
        "\nCase:",
        record[
            "golden_case_id"
        ]
    )

    print(
        "Rank:",
        record[
            "selection_rank"
        ]
    )

    print(
        "Population Unit:",
        record[
            "candidate_identity"
        ][
            "population_unit_id"
        ]
    )

    print(
        "Heading:",
        record[
            "candidate_snapshot"
        ][
            "heading_path"
        ]
    )

    print(
        "Evidence documents:",
        record[
            "evidence_document_count"
        ]
    )


print(
    "\n=== CODE CELL 52 GATES ==="
)


for name, passed in (
    cell_52_gates.items()
):

    print(
        name,
        ":",
        "PASS"
        if passed
        else "FAIL"
    )


print(
    "\nCODE CELL 52 GATE:",
    "PASS"
    if cell_52_pass
    else "FAIL"
)


assert cell_52_pass, (
    "Exercise-unit V2 Golden Case staging is still not valid. "
    "Do not advance or supersede anything. Repair this SAME "
    "Code Cell 52 using the exact failing canonical-evidence "
    "or staging gate reported above."
)
```

```python
# Code Cell 53

# ============================================================
# Legacy -> V2 Supersession Lineage Plan
#
# INPUTS:
#   - 38 legacy Exercise Golden Cases
#   - legacy -> structural-unit traceability
#   - rebuilt selected 38 Exercise Units
#   - staged 38 V2 Golden Cases
#
# PURPOSE:
#   classify every old and new case BEFORE activation.
#
# NO active registry mutation.
# NO legacy deletion.
# NO V2 activation.
# NO Task rebuild.
# NO Target rebuild.
# NO Pilot rebuild.
# NO question generation.
# NO artifact writes.
# ============================================================


# ============================================================
# 1) Snapshot active state
# ============================================================

cell53_golden_snapshot_before = hashlib.sha256(
    stable_json_key(
        golden_case_records
    ).encode(
        "utf-8"
    )
).hexdigest()


cell53_auto_snapshot_before = hashlib.sha256(
    stable_json_key(
        auto_case_records
    ).encode(
        "utf-8"
    )
).hexdigest()


cell53_task_snapshot_before = hashlib.sha256(
    stable_json_key(
        task_draft_plan_rows
    ).encode(
        "utf-8"
    )
).hexdigest()


cell53_target_snapshot_before = hashlib.sha256(
    stable_json_key(
        target_evidence_records
    ).encode(
        "utf-8"
    )
).hexdigest()


cell53_pilot_snapshot_before = hashlib.sha256(
    stable_json_key(
        pilot_selection_rows
    ).encode(
        "utf-8"
    )
).hexdigest()


cell53_staging_snapshot_before = hashlib.sha256(
    stable_json_key(
        staged_exercise_v2_golden_case_records
    ).encode(
        "utf-8"
    )
).hexdigest()


# ============================================================
# 2) Inspect actual legacy mapping schema
# ============================================================

cell53_legacy_mapping_key_union = sorted(
    set().union(
        *[
            set(
                row.keys()
            )
            for row
            in legacy_38_to_population_mapping
        ]
    )
)


cell53_legacy_mapping_key_intersection = sorted(
    set.intersection(
        *[
            set(
                row.keys()
            )
            for row
            in legacy_38_to_population_mapping
        ]
    )
)


# ============================================================
# 3) Resolve population_unit_id for each legacy mapping row
#
# Use only already-established traceability.
# Do not infer from text/headings.
# ============================================================

cell53_proposed_case_to_population_lookup = (
    globals().get(
        "proposed_case_id_to_population_id"
    )
)


def cell53_resolve_population_unit_id(
    mapping_row,
):

    direct_field_candidates = [
        "population_unit_id",
        "exercise_population_unit_id",
        "proposed_population_unit_id",
    ]


    direct_values = [
        mapping_row.get(
            field_name
        )

        for field_name
        in direct_field_candidates

        if (
            mapping_row.get(
                field_name
            )
            is not None
        )
    ]


    direct_values = list(
        dict.fromkeys(
            direct_values
        )
    )


    if (
        len(
            direct_values
        )
        > 1
    ):

        raise AssertionError(
            "Conflicting direct population-unit IDs in "
            f"legacy mapping row: {mapping_row}"
        )


    if (
        len(
            direct_values
        )
        == 1
    ):

        return {
            "population_unit_id":
                direct_values[
                    0
                ],

            "resolution_method":
                "DIRECT_MAPPING_FIELD",
        }


    proposed_case_id = (
        mapping_row.get(
            "proposed_exercise_unit_case_id"
        )
    )


    if (
        proposed_case_id
        is not None

        and

        isinstance(
            cell53_proposed_case_to_population_lookup,
            dict,
        )

        and

        proposed_case_id
        in
        cell53_proposed_case_to_population_lookup
    ):

        return {
            "population_unit_id":
                cell53_proposed_case_to_population_lookup[
                    proposed_case_id
                ],

            "resolution_method":
                "PROPOSED_CASE_TO_POPULATION_TRACEABILITY",
        }


    raise AssertionError(
        "Could not resolve population_unit_id from existing "
        f"traceability row: {mapping_row}"
    )


# ============================================================
# 4) Build normalized 38-row legacy lineage table
# ============================================================

cell53_legacy_lineage_rows = []

cell53_legacy_lineage_resolution_failures = []


for mapping_row in (
    legacy_38_to_population_mapping
):

    legacy_golden_case_id = (
        mapping_row.get(
            "legacy_golden_case_id"
        )
    )


    if (
        legacy_golden_case_id
        is None
    ):

        cell53_legacy_lineage_resolution_failures.append(
            {
                "reason":
                    "MISSING_LEGACY_GOLDEN_CASE_ID",

                "mapping_row":
                    mapping_row,
            }
        )

        continue


    try:

        resolved_population = (
            cell53_resolve_population_unit_id(
                mapping_row
            )
        )


    except Exception as exc:

        cell53_legacy_lineage_resolution_failures.append(
            {
                "legacy_golden_case_id":
                    legacy_golden_case_id,

                "error_type":
                    type(
                        exc
                    ).__name__,

                "error_message":
                    str(
                        exc
                    ),
            }
        )

        continue


    population_unit_id = (
        resolved_population[
            "population_unit_id"
        ]
    )


    cell53_legacy_lineage_rows.append(
        {
            "legacy_golden_case_id":
                legacy_golden_case_id,

            "population_unit_id":
                population_unit_id,

            "resolution_method":
                resolved_population[
                    "resolution_method"
                ],

            "legacy_candidate_document_id":
                mapping_row.get(
                    "legacy_candidate_document_id"
                ),

            "legacy_candidate_position":
                mapping_row.get(
                    "legacy_candidate_position"
                ),

            "legacy_relationship":
                mapping_row.get(
                    "relationship"
                ),

            "proposed_exercise_unit_case_id":
                mapping_row.get(
                    "proposed_exercise_unit_case_id"
                ),
        }
    )


# ============================================================
# 5) Validate all legacy IDs against active Golden Cases
# ============================================================

cell53_missing_legacy_active_cases = sorted(
    {
        row[
            "legacy_golden_case_id"
        ]

        for row
        in cell53_legacy_lineage_rows
    }
    -
    set(
        golden_case_records_by_id.keys()
    )
)


# ============================================================
# 6) Group legacy cases by structural Exercise Unit
# ============================================================

cell53_legacy_case_ids_by_population_unit = defaultdict(
    list
)


for row in (
    cell53_legacy_lineage_rows
):

    cell53_legacy_case_ids_by_population_unit[
        row[
            "population_unit_id"
        ]
    ].append(
        row[
            "legacy_golden_case_id"
        ]
    )


for population_unit_id in list(
    cell53_legacy_case_ids_by_population_unit.keys()
):

    cell53_legacy_case_ids_by_population_unit[
        population_unit_id
    ] = sorted(
        cell53_legacy_case_ids_by_population_unit[
            population_unit_id
        ]
    )


cell53_legacy_represented_structural_units = set(
    cell53_legacy_case_ids_by_population_unit.keys()
)


# ============================================================
# 7) Build staged V2 lookup by structural Exercise Unit
# ============================================================

cell53_staged_case_by_population_unit = {}


cell53_duplicate_staged_population_units = []


for staged_case in (
    staged_exercise_v2_golden_case_records
):

    population_unit_id = (
        staged_case[
            "candidate_identity"
        ][
            "population_unit_id"
        ]
    )


    if (
        population_unit_id
        in
        cell53_staged_case_by_population_unit
    ):

        cell53_duplicate_staged_population_units.append(
            population_unit_id
        )

        continue


    cell53_staged_case_by_population_unit[
        population_unit_id
    ] = staged_case


cell53_staged_population_unit_ids = set(
    cell53_staged_case_by_population_unit.keys()
)


# ============================================================
# 8) Structural-unit classification
# ============================================================

cell53_retained_legacy_structural_units = (
    cell53_legacy_represented_structural_units
    &
    set(
        selected_exercise_population_unit_ids
    )
)


cell53_excluded_legacy_structural_units = (
    cell53_legacy_represented_structural_units
    -
    set(
        selected_exercise_population_unit_ids
    )
)


cell53_newly_admitted_structural_units = (
    set(
        selected_exercise_population_unit_ids
    )
    -
    cell53_legacy_represented_structural_units
)


# ============================================================
# 9) Build legacy-case Supersession Plan
# ============================================================

cell53_legacy_case_supersession_rows = []


for lineage_row in sorted(
    cell53_legacy_lineage_rows,
    key=lambda row:
        row[
            "legacy_golden_case_id"
        ],
):

    legacy_case_id = (
        lineage_row[
            "legacy_golden_case_id"
        ]
    )


    population_unit_id = (
        lineage_row[
            "population_unit_id"
        ]
    )


    if (
        population_unit_id
        in
        cell53_retained_legacy_structural_units
    ):

        staged_successor = (
            cell53_staged_case_by_population_unit.get(
                population_unit_id
            )
        )


        successor_case_id = (
            staged_successor[
                "golden_case_id"
            ]
            if staged_successor
            is not None
            else None
        )


        disposition = (
            "SUPERSEDE_BY_V2_CASE_FOR_SAME_STRUCTURAL_UNIT"
        )


    elif (
        population_unit_id
        in
        cell53_excluded_legacy_structural_units
    ):

        successor_case_id = None


        disposition = (
            "RETIRE_NO_SUCCESSOR_UNIT_EXCLUDED_FROM_REBUILT_SAMPLE"
        )


    else:

        successor_case_id = None


        disposition = (
            "UNRESOLVED_STRUCTURAL_DISPOSITION"
        )


    cell53_legacy_case_supersession_rows.append(
        {
            **lineage_row,

            "legacy_case_disposition":
                disposition,

            "successor_v2_golden_case_id":
                successor_case_id,

            "activation_status":
                "NOT_APPLIED",
        }
    )


# ============================================================
# 10) Build staged V2 predecessor / admission plan
# ============================================================

cell53_v2_case_lineage_rows = []


for staged_case in (
    staged_exercise_v2_golden_case_records
):

    population_unit_id = (
        staged_case[
            "candidate_identity"
        ][
            "population_unit_id"
        ]
    )


    predecessor_legacy_ids = sorted(
        cell53_legacy_case_ids_by_population_unit.get(
            population_unit_id,
            [],
        )
    )


    if (
        predecessor_legacy_ids
    ):

        lineage_class = (
            "V2_SUCCESSOR_OF_LEGACY_STRUCTURAL_UNIT"
        )


    else:

        lineage_class = (
            "NEW_V2_CASE_FROM_FULL_CORPUS_DISCOVERY"
        )


    cell53_v2_case_lineage_rows.append(
        {
            "v2_golden_case_id":
                staged_case[
                    "golden_case_id"
                ],

            "population_unit_id":
                population_unit_id,

            "selection_rank":
                staged_case[
                    "selection_rank"
                ],

            "heading_path":
                staged_case[
                    "candidate_snapshot"
                ][
                    "heading_path"
                ],

            "predecessor_legacy_golden_case_ids":
                predecessor_legacy_ids,

            "predecessor_count":
                len(
                    predecessor_legacy_ids
                ),

            "lineage_class":
                lineage_class,

            "activation_status":
                "NOT_APPLIED",
        }
    )


# ============================================================
# 11) Many-to-one consolidation audit
# ============================================================

cell53_predecessor_count_distribution = Counter(
    row[
        "predecessor_count"
    ]

    for row
    in cell53_v2_case_lineage_rows
)


cell53_many_to_one_successor_rows = [
    row

    for row
    in cell53_v2_case_lineage_rows

    if (
        row[
            "predecessor_count"
        ]
        > 1
    )
]


cell53_single_predecessor_successor_rows = [
    row

    for row
    in cell53_v2_case_lineage_rows

    if (
        row[
            "predecessor_count"
        ]
        == 1
    )
]


cell53_new_admission_rows = [
    row

    for row
    in cell53_v2_case_lineage_rows

    if (
        row[
            "predecessor_count"
        ]
        == 0
    )
]


# ============================================================
# 12) Legacy disposition counts
# ============================================================

cell53_legacy_disposition_counts = Counter(
    row[
        "legacy_case_disposition"
    ]

    for row
    in cell53_legacy_case_supersession_rows
)


cell53_legacy_cases_with_successor = [
    row

    for row
    in cell53_legacy_case_supersession_rows

    if (
        row[
            "successor_v2_golden_case_id"
        ]
        is not None
    )
]


cell53_legacy_cases_retired_without_successor = [
    row

    for row
    in cell53_legacy_case_supersession_rows

    if (
        row[
            "legacy_case_disposition"
        ]
        ==
        "RETIRE_NO_SUCCESSOR_UNIT_EXCLUDED_FROM_REBUILT_SAMPLE"
    )
]


# ============================================================
# 13) Confirm successor coverage at structural-unit level
# ============================================================

cell53_retained_units_without_staged_successor = sorted(
    cell53_retained_legacy_structural_units
    -
    cell53_staged_population_unit_ids
)


cell53_excluded_units_with_unexpected_staged_successor = sorted(
    cell53_excluded_legacy_structural_units
    &
    cell53_staged_population_unit_ids
)


cell53_new_units_missing_staged_case = sorted(
    cell53_newly_admitted_structural_units
    -
    cell53_staged_population_unit_ids
)


# ============================================================
# 14) Old failed Exercise Pilot lineage
# ============================================================

cell53_failed_pilot_supersession_row = next(
    (
        row

        for row
        in cell53_legacy_case_supersession_rows

        if (
            row[
                "legacy_golden_case_id"
            ]
            ==
            CELL46_FAILED_EXERCISE_PILOT_LEGACY_CASE_ID
        )
    ),
    None,
)


# ============================================================
# 15) Downstream dependency invalidation plan
#
# All 38 legacy case IDs change/disappear from active Case set.
# Therefore Task and Target objects tied to those IDs cannot
# simply be retained.
#
# No rebuild performed here.
# ============================================================

cell53_legacy_task_ids = {
    task_id

    for legacy_case_id
    in cell46_legacy_exercise_case_ids

    for task_id
    in cell47_task_draft_ids_by_case[
        legacy_case_id
    ]
}


cell53_legacy_target_ids = {
    target_id

    for legacy_case_id
    in cell46_legacy_exercise_case_ids

    for target_id
    in cell47_target_ids_by_case[
        legacy_case_id
    ]
}


cell53_downstream_rebuild_plan = {
    "legacy_case_records":
        {
            "affected":
                38,

            "action":
                (
                    "ATOMICALLY_REPLACE_ACTIVE_EXERCISE_SLICE_"
                    "ONLY_AFTER_DOWNSTREAM_STAGING"
                ),
        },

    "legacy_task_drafts":
        {
            "affected":
                len(
                    cell53_legacy_task_ids
                ),

            "action":
                "REBUILD_FOR_38_V2_CASE_IDS",
        },

    "legacy_target_evidence":
        {
            "affected":
                len(
                    cell53_legacy_target_ids
                ),

            "action":
                "REBUILD_FROM_V2_EXERCISE_UNIT_EVIDENCE",
        },

    "exercise_pilot":
        {
            "affected_legacy_pilot_case_count":
                len(
                    cell47_pilot_legacy_case_ids
                ),

            "action":
                (
                    "RESELECT_AFTER_V2_TARGET_REBUILD_"
                    "DO_NOT_CARRY_FORWARD_OLD_PILOT"
                ),
        },

    "historical_audit_structures":
        {
            "action":
                "PRESERVE_UNMODIFIED_FOR_TRACEABILITY",
        },

    "question_generation":
        {
            "action":
                "REMAINS_NOT_STARTED",
        },
}


# ============================================================
# 16) Supersession plan snapshot
# ============================================================

cell53_supersession_snapshot_payload = {
    "legacy_case_supersession":
        [
            {
                "legacy_golden_case_id":
                    row[
                        "legacy_golden_case_id"
                    ],

                "population_unit_id":
                    row[
                        "population_unit_id"
                    ],

                "disposition":
                    row[
                        "legacy_case_disposition"
                    ],

                "successor_v2_golden_case_id":
                    row[
                        "successor_v2_golden_case_id"
                    ],
            }

            for row
            in cell53_legacy_case_supersession_rows
        ],

    "v2_case_lineage":
        [
            {
                "v2_golden_case_id":
                    row[
                        "v2_golden_case_id"
                    ],

                "population_unit_id":
                    row[
                        "population_unit_id"
                    ],

                "predecessor_legacy_golden_case_ids":
                    row[
                        "predecessor_legacy_golden_case_ids"
                    ],

                "lineage_class":
                    row[
                        "lineage_class"
                    ],
            }

            for row
            in cell53_v2_case_lineage_rows
        ],
}


cell53_supersession_plan_snapshot_sha256 = hashlib.sha256(
    stable_json_key(
        cell53_supersession_snapshot_payload
    ).encode(
        "utf-8"
    )
).hexdigest()


# ============================================================
# 17) Plan registry
# ============================================================

EXERCISE_V2_SUPERSESSION_LINEAGE_PLAN_V1_DRAFT = {
    "version":
        "exercise_v2_supersession_lineage_plan_v1_draft",

    "status":
        "PLANNED_AND_VALIDATED_NOT_APPLIED",

    "legacy_case_count":
        len(
            cell53_legacy_case_supersession_rows
        ),

    "legacy_structural_unit_count":
        len(
            cell53_legacy_represented_structural_units
        ),

    "retained_legacy_structural_unit_count":
        len(
            cell53_retained_legacy_structural_units
        ),

    "excluded_legacy_structural_unit_count":
        len(
            cell53_excluded_legacy_structural_units
        ),

    "newly_admitted_structural_unit_count":
        len(
            cell53_newly_admitted_structural_units
        ),

    "staged_v2_case_count":
        len(
            cell53_v2_case_lineage_rows
        ),

    "supersession_plan_snapshot_sha256":
        cell53_supersession_plan_snapshot_sha256,

    "activation_status":
        "NOT_STARTED",

    "active_case_registry_mutation":
        False,

    "task_rebuild_status":
        "NOT_STARTED",

    "target_rebuild_status":
        "NOT_STARTED",

    "pilot_rebuild_status":
        "NOT_STARTED",

    "question_generation_status":
        "NOT_STARTED",

    "artifact_write_status":
        "NONE",
}


# ============================================================
# 18) Independent no-mutation recheck
# ============================================================

cell53_golden_snapshot_after = hashlib.sha256(
    stable_json_key(
        golden_case_records
    ).encode(
        "utf-8"
    )
).hexdigest()


cell53_auto_snapshot_after = hashlib.sha256(
    stable_json_key(
        auto_case_records
    ).encode(
        "utf-8"
    )
).hexdigest()


cell53_task_snapshot_after = hashlib.sha256(
    stable_json_key(
        task_draft_plan_rows
    ).encode(
        "utf-8"
    )
).hexdigest()


cell53_target_snapshot_after = hashlib.sha256(
    stable_json_key(
        target_evidence_records
    ).encode(
        "utf-8"
    )
).hexdigest()


cell53_pilot_snapshot_after = hashlib.sha256(
    stable_json_key(
        pilot_selection_rows
    ).encode(
        "utf-8"
    )
).hexdigest()


cell53_staging_snapshot_after = hashlib.sha256(
    stable_json_key(
        staged_exercise_v2_golden_case_records
    ).encode(
        "utf-8"
    )
).hexdigest()


# ============================================================
# 19) Validation gates
# ============================================================

exact_38_legacy_mapping_rows_gate = (
    len(
        legacy_38_to_population_mapping
    )
    == 38

    and

    len(
        cell53_legacy_lineage_rows
    )
    == 38

    and

    len(
        cell53_legacy_lineage_resolution_failures
    )
    == 0
)


all_38_legacy_case_ids_unique_gate = (
    len(
        {
            row[
                "legacy_golden_case_id"
            ]

            for row
            in cell53_legacy_lineage_rows
        }
    )
    == 38
)


all_38_legacy_cases_exist_active_gate = (
    len(
        cell53_missing_legacy_active_cases
    )
    == 0
)


exact_18_legacy_structural_units_gate = (
    len(
        cell53_legacy_represented_structural_units
    )
    == 18
)


exact_15_retained_legacy_units_gate = (
    len(
        cell53_retained_legacy_structural_units
    )
    == 15
)


exact_3_excluded_legacy_units_gate = (
    len(
        cell53_excluded_legacy_structural_units
    )
    == 3
)


exact_23_newly_admitted_units_gate = (
    len(
        cell53_newly_admitted_structural_units
    )
    == 23
)


exact_38_staged_v2_cases_gate = (
    len(
        cell53_staged_case_by_population_unit
    )
    == 38

    and

    len(
        cell53_duplicate_staged_population_units
    )
    == 0
)


retained_units_have_successors_gate = (
    len(
        cell53_retained_units_without_staged_successor
    )
    == 0
)


excluded_units_have_no_successor_gate = (
    len(
        cell53_excluded_units_with_unexpected_staged_successor
    )
    == 0
)


new_units_have_staged_cases_gate = (
    len(
        cell53_new_units_missing_staged_case
    )
    == 0
)


all_legacy_dispositions_resolved_gate = all(
    row[
        "legacy_case_disposition"
    ]
    in {
        "SUPERSEDE_BY_V2_CASE_FOR_SAME_STRUCTURAL_UNIT",
        "RETIRE_NO_SUCCESSOR_UNIT_EXCLUDED_FROM_REBUILT_SAMPLE",
    }

    for row
    in cell53_legacy_case_supersession_rows
)


all_v2_lineages_resolved_gate = all(
    row[
        "lineage_class"
    ]
    in {
        "V2_SUCCESSOR_OF_LEGACY_STRUCTURAL_UNIT",
        "NEW_V2_CASE_FROM_FULL_CORPUS_DISCOVERY",
    }

    for row
    in cell53_v2_case_lineage_rows
)


exact_23_v2_new_admissions_gate = (
    len(
        cell53_new_admission_rows
    )
    == 23
)


legacy_task_dependency_count_gate = (
    len(
        cell53_legacy_task_ids
    )
    == 38
)


legacy_target_dependency_count_gate = (
    len(
        cell53_legacy_target_ids
    )
    == 38
)


failed_pilot_lineage_resolved_gate = (
    cell53_failed_pilot_supersession_row
    is not None

    and

    cell53_failed_pilot_supersession_row[
        "legacy_case_disposition"
    ]
    in {
        "SUPERSEDE_BY_V2_CASE_FOR_SAME_STRUCTURAL_UNIT",
        "RETIRE_NO_SUCCESSOR_UNIT_EXCLUDED_FROM_REBUILT_SAMPLE",
    }
)


active_registries_unchanged_gate = (
    cell53_golden_snapshot_before
    ==
    cell53_golden_snapshot_after

    and

    cell53_auto_snapshot_before
    ==
    cell53_auto_snapshot_after

    and

    cell53_task_snapshot_before
    ==
    cell53_task_snapshot_after

    and

    cell53_target_snapshot_before
    ==
    cell53_target_snapshot_after

    and

    cell53_pilot_snapshot_before
    ==
    cell53_pilot_snapshot_after

    and

    cell53_staging_snapshot_before
    ==
    cell53_staging_snapshot_after

    and

    len(
        golden_case_records
    )
    == 724

    and

    len(
        auto_case_records
    )
    == 572

    and

    len(
        task_draft_plan_rows
    )
    == 724

    and

    len(
        target_evidence_records
    )
    == 572

    and

    len(
        pilot_selection_rows
    )
    == 23
)


activation_not_started_gate = (
    EXERCISE_V2_SUPERSESSION_LINEAGE_PLAN_V1_DRAFT[
        "activation_status"
    ]
    ==
    "NOT_STARTED"
)


task_rebuild_not_started_gate = (
    EXERCISE_V2_SUPERSESSION_LINEAGE_PLAN_V1_DRAFT[
        "task_rebuild_status"
    ]
    ==
    "NOT_STARTED"
)


target_rebuild_not_started_gate = (
    EXERCISE_V2_SUPERSESSION_LINEAGE_PLAN_V1_DRAFT[
        "target_rebuild_status"
    ]
    ==
    "NOT_STARTED"
)


pilot_rebuild_not_started_gate = (
    EXERCISE_V2_SUPERSESSION_LINEAGE_PLAN_V1_DRAFT[
        "pilot_rebuild_status"
    ]
    ==
    "NOT_STARTED"
)


question_generation_not_started_gate = (
    EXERCISE_V2_SUPERSESSION_LINEAGE_PLAN_V1_DRAFT[
        "question_generation_status"
    ]
    ==
    "NOT_STARTED"
)


no_artifact_write_gate = (
    EXERCISE_V2_SUPERSESSION_LINEAGE_PLAN_V1_DRAFT[
        "artifact_write_status"
    ]
    ==
    "NONE"
)


cell_53_gates = {
    "exact_38_legacy_traceability_rows_resolved":
        exact_38_legacy_mapping_rows_gate,

    "all_38_legacy_golden_case_ids_unique":
        all_38_legacy_case_ids_unique_gate,

    "all_38_legacy_cases_exist_in_current_active_registry":
        all_38_legacy_cases_exist_active_gate,

    "legacy_38_cases_collapse_to_exact_18_structural_units":
        exact_18_legacy_structural_units_gate,

    "exact_15_legacy_structural_units_retained_in_rebuilt_sample":
        exact_15_retained_legacy_units_gate,

    "exact_3_legacy_structural_units_excluded_from_rebuilt_sample":
        exact_3_excluded_legacy_units_gate,

    "exact_23_selected_units_are_newly_admitted":
        exact_23_newly_admitted_units_gate,

    "exact_38_unique_staged_v2_cases_available":
        exact_38_staged_v2_cases_gate,

    "every_retained_legacy_unit_has_exact_staged_successor":
        retained_units_have_successors_gate,

    "excluded_legacy_units_have_no_unexpected_v2_successor":
        excluded_units_have_no_successor_gate,

    "every_newly_admitted_selected_unit_has_staged_v2_case":
        new_units_have_staged_cases_gate,

    "all_38_legacy_case_dispositions_resolved":
        all_legacy_dispositions_resolved_gate,

    "all_38_v2_case_lineage_classes_resolved":
        all_v2_lineages_resolved_gate,

    "exact_23_v2_cases_have_no_legacy_predecessor":
        exact_23_v2_new_admissions_gate,

    "exact_38_legacy_task_draft_ids_require_rebuild":
        legacy_task_dependency_count_gate,

    "exact_38_legacy_target_evidence_ids_require_rebuild":
        legacy_target_dependency_count_gate,

    "failed_exercise_pilot_legacy_case_lineage_resolved":
        failed_pilot_lineage_resolved_gate,

    "all_active_and_staging_registries_content_unchanged":
        active_registries_unchanged_gate,

    "activation_not_started":
        activation_not_started_gate,

    "task_rebuild_not_started":
        task_rebuild_not_started_gate,

    "target_rebuild_not_started":
        target_rebuild_not_started_gate,

    "pilot_rebuild_not_started":
        pilot_rebuild_not_started_gate,

    "question_generation_not_started":
        question_generation_not_started_gate,

    "no_artifact_write":
        no_artifact_write_gate,
}


cell_53_pass = all(
    cell_53_gates.values()
)


# ============================================================
# 20) Output
# ============================================================

print(
    "=== LEGACY -> V2 SUPERSESSION LINEAGE PLAN ==="
)


print(
    "Legacy mapping schema intersection:",
    cell53_legacy_mapping_key_intersection
)


print(
    "Legacy mapping schema union:",
    cell53_legacy_mapping_key_union
)


print(
    "\n=== STRUCTURAL UNIT ACCOUNTING ==="
)


print(
    "Legacy Golden Cases:",
    len(
        cell53_legacy_lineage_rows
    )
)


print(
    "Legacy represented structural units:",
    len(
        cell53_legacy_represented_structural_units
    )
)


print(
    "Retained legacy structural units:",
    len(
        cell53_retained_legacy_structural_units
    )
)


print(
    "Excluded legacy structural units:",
    len(
        cell53_excluded_legacy_structural_units
    )
)


print(
    "Newly admitted selected structural units:",
    len(
        cell53_newly_admitted_structural_units
    )
)


print(
    "Staged V2 structural units:",
    len(
        cell53_staged_population_unit_ids
    )
)


print(
    "\n=== LEGACY CASE DISPOSITIONS ==="
)


for disposition, count in sorted(
    cell53_legacy_disposition_counts.items()
):

    print(
        "-",
        disposition,
        ":",
        count
    )


print(
    "\nLegacy cases with staged successor:",
    len(
        cell53_legacy_cases_with_successor
    )
)


print(
    "Legacy cases retired without successor:",
    len(
        cell53_legacy_cases_retired_without_successor
    )
)


print(
    "\n=== V2 PREDECESSOR DISTRIBUTION ==="
)


for predecessor_count, case_count in sorted(
    cell53_predecessor_count_distribution.items()
):

    print(
        "- predecessor legacy cases:",
        predecessor_count,
        "| V2 cases:",
        case_count,
    )


print(
    "Many-to-one V2 successor cases:",
    len(
        cell53_many_to_one_successor_rows
    )
)


print(
    "Single-predecessor V2 successor cases:",
    len(
        cell53_single_predecessor_successor_rows
    )
)


print(
    "New V2 admissions with no predecessor:",
    len(
        cell53_new_admission_rows
    )
)


print(
    "\n=== MANY-TO-ONE CONSOLIDATIONS ==="
)


if (
    cell53_many_to_one_successor_rows
):

    for row in (
        cell53_many_to_one_successor_rows
    ):

        print(
            "\nV2 Case:",
            row[
                "v2_golden_case_id"
            ]
        )

        print(
            "Population Unit:",
            row[
                "population_unit_id"
            ]
        )

        print(
            "Heading:",
            row[
                "heading_path"
            ]
        )

        print(
            "Legacy predecessors:",
            row[
                "predecessor_count"
            ]
        )

        for legacy_case_id in (
            row[
                "predecessor_legacy_golden_case_ids"
            ]
        ):

            print(
                "  -",
                legacy_case_id
            )

else:

    print(
        "NONE"
    )


print(
    "\n=== LEGACY RETIREMENTS WITHOUT SUCCESSOR ==="
)


if (
    cell53_legacy_cases_retired_without_successor
):

    for row in (
        cell53_legacy_cases_retired_without_successor
    ):

        print(
            "-",
            row[
                "legacy_golden_case_id"
            ],
            "| unit:",
            row[
                "population_unit_id"
            ],
            "| document:",
            row[
                "legacy_candidate_document_id"
            ],
        )

else:

    print(
        "NONE"
    )


print(
    "\n=== NEWLY ADMITTED V2 CASES ==="
)


for row in (
    cell53_new_admission_rows
):

    print(
        "-",
        row[
            "v2_golden_case_id"
        ],
        "| unit:",
        row[
            "population_unit_id"
        ],
        "| heading:",
        row[
            "heading_path"
        ],
    )


print(
    "\n=== FAILED EXERCISE PILOT LINEAGE ==="
)


print(
    cell53_failed_pilot_supersession_row
)


print(
    "\n=== DOWNSTREAM REBUILD PLAN ==="
)


for dependency_name, plan in (
    cell53_downstream_rebuild_plan.items()
):

    print(
        "-",
        dependency_name,
        ":",
        plan,
    )


print(
    "\nSupersession Plan Snapshot SHA256:",
    cell53_supersession_plan_snapshot_sha256
)


print(
    "\n=== ACTIVE REGISTRY COUNTS ==="
)


print(
    "Golden:",
    len(
        golden_case_records
    )
)


print(
    "Auto:",
    len(
        auto_case_records
    )
)


print(
    "Task plans:",
    len(
        task_draft_plan_rows
    )
)


print(
    "Targets:",
    len(
        target_evidence_records
    )
)


print(
    "Pilots:",
    len(
        pilot_selection_rows
    )
)


print(
    "\n=== CODE CELL 53 GATES ==="
)


for name, passed in (
    cell_53_gates.items()
):

    print(
        name,
        ":",
        "PASS"
        if passed
        else "FAIL"
    )


print(
    "\nCODE CELL 53 GATE:",
    "PASS"
    if cell_53_pass
    else "FAIL"
)


assert cell_53_pass, (
    "Legacy -> V2 Supersession Lineage Plan failed. "
    "Do not activate staged V2 cases or mutate any downstream "
    "registry until all 38 legacy dispositions, all 38 V2 "
    "lineages, structural-unit accounting and Task/Target "
    "dependency scope are completely resolved."
)
```

```python
# Code Cell 54

# ============================================================
# Legacy Exercise Task/Plan Contract & Stable-ID Discovery
#
# Goal:
#   discover the exact Task/Plan contract used by the 38
#   legacy Exercise Golden Cases before V2 Task staging.
#
# READ-ONLY.
#
# NO V2 Task creation.
# NO Case activation.
# NO supersession application.
# NO Target rebuild.
# NO Pilot rebuild.
# NO question generation.
# NO artifact writes.
# ============================================================


# ============================================================
# 1) Snapshot namespace and active registries
# ============================================================

cell54_globals_snapshot = dict(
    globals()
)

cell54_globals_snapshot_items = list(
    cell54_globals_snapshot.items()
)


cell54_task_snapshot_before = hashlib.sha256(
    stable_json_key(
        task_draft_plan_rows
    ).encode(
        "utf-8"
    )
).hexdigest()


cell54_ready_plan_snapshot_before = hashlib.sha256(
    stable_json_key(
        ready_case_plans
    ).encode(
        "utf-8"
    )
).hexdigest()


cell54_design_snapshot_before = hashlib.sha256(
    stable_json_key(
        case_task_design
    ).encode(
        "utf-8"
    )
).hexdigest()


cell54_groundability_snapshot_before = hashlib.sha256(
    stable_json_key(
        case_groundability_precheck
    ).encode(
        "utf-8"
    )
).hexdigest()


cell54_target_snapshot_before = hashlib.sha256(
    stable_json_key(
        target_evidence_records
    ).encode(
        "utf-8"
    )
).hexdigest()


cell54_pilot_snapshot_before = hashlib.sha256(
    stable_json_key(
        pilot_selection_rows
    ).encode(
        "utf-8"
    )
).hexdigest()


cell54_golden_snapshot_before = hashlib.sha256(
    stable_json_key(
        golden_case_records
    ).encode(
        "utf-8"
    )
).hexdigest()


cell54_staged_case_snapshot_before = hashlib.sha256(
    stable_json_key(
        staged_exercise_v2_golden_case_records
    ).encode(
        "utf-8"
    )
).hexdigest()


# ============================================================
# 2) Exact 38 legacy Exercise Task rows
# ============================================================

cell54_legacy_task_rows = sorted(
    [
        row

        for row
        in task_draft_plan_rows

        if (
            row[
                "golden_case_id"
            ]
            in
            cell46_legacy_exercise_case_ids
        )
    ],
    key=lambda row:
        row[
            "golden_case_id"
        ],
)


cell54_legacy_ready_case_plan_rows = sorted(
    [
        row

        for row
        in ready_case_plans

        if (
            row[
                "golden_case_id"
            ]
            in
            cell46_legacy_exercise_case_ids
        )
    ],
    key=lambda row:
        row[
            "golden_case_id"
        ],
)


# ============================================================
# 3) Robust case-record resolver for dict registries
# ============================================================

def cell54_resolve_case_registry_record(
    registry,
    golden_case_id,
):

    if not isinstance(
        registry,
        dict,
    ):

        return None


    if (
        golden_case_id
        in
        registry
    ):

        record = registry[
            golden_case_id
        ]


        if isinstance(
            record,
            dict,
        ):

            return record


    matching_records = [
        value

        for value
        in registry.values()

        if (
            isinstance(
                value,
                dict,
            )

            and

            value.get(
                "golden_case_id"
            )
            ==
            golden_case_id
        )
    ]


    if (
        len(
            matching_records
        )
        == 1
    ):

        return matching_records[
            0
        ]


    return None


# ============================================================
# 4) Groundability + Task Design for all 38 legacy cases
# ============================================================

cell54_legacy_groundability_rows = []

cell54_legacy_task_design_rows = []

cell54_missing_groundability_cases = []

cell54_missing_task_design_cases = []


for legacy_case_id in sorted(
    cell46_legacy_exercise_case_ids
):

    groundability_record = (
        cell54_resolve_case_registry_record(
            case_groundability_precheck,
            legacy_case_id,
        )
    )


    task_design_record = (
        cell54_resolve_case_registry_record(
            case_task_design,
            legacy_case_id,
        )
    )


    if (
        groundability_record
        is None
    ):

        cell54_missing_groundability_cases.append(
            legacy_case_id
        )

    else:

        cell54_legacy_groundability_rows.append(
            groundability_record
        )


    if (
        task_design_record
        is None
    ):

        cell54_missing_task_design_cases.append(
            legacy_case_id
        )

    else:

        cell54_legacy_task_design_rows.append(
            task_design_record
        )


# ============================================================
# 5) Task Draft schema audit
# ============================================================

cell54_task_key_sets = [
    set(
        row.keys()
    )

    for row
    in cell54_legacy_task_rows
]


cell54_task_key_union = (
    set().union(
        *cell54_task_key_sets
    )
    if cell54_task_key_sets
    else set()
)


cell54_task_key_intersection = (
    set(
        cell54_task_key_sets[
            0
        ]
    )
    if cell54_task_key_sets
    else set()
)


for key_set in (
    cell54_task_key_sets[
        1:
    ]
):

    cell54_task_key_intersection &= (
        key_set
    )


cell54_task_schema_uniform = (
    cell54_task_key_union
    ==
    cell54_task_key_intersection
)


# ============================================================
# 6) Field-value distributions
# ============================================================

cell54_task_field_distinct_counts = {}

cell54_task_constant_fields = {}


for field_name in sorted(
    cell54_task_key_union
):

    values = [
        row.get(
            field_name
        )

        for row
        in cell54_legacy_task_rows
    ]


    serialized_values = [
        stable_json_key(
            value
        )

        for value
        in values
    ]


    distinct_values = set(
        serialized_values
    )


    cell54_task_field_distinct_counts[
        field_name
    ] = len(
        distinct_values
    )


    if (
        len(
            distinct_values
        )
        == 1
    ):

        cell54_task_constant_fields[
            field_name
        ] = values[
            0
        ]


# ============================================================
# 7) High-value semantic distributions
# ============================================================

CELL54_SEMANTIC_TASK_FIELDS = [
    "task_family",
    "question_archetype",
    "evidence_target_method",
    "evidence_scope",
    "authority_expectation",
    "context_expectation",
    "construction_eligibility",
    "target_evidence_status",
    "citation_target_status",
    "question_text_status",
    "expected_answer_status",
    "required_dimensions",
]


cell54_semantic_field_distributions = {}


for field_name in (
    CELL54_SEMANTIC_TASK_FIELDS
):

    if (
        field_name
        not in
        cell54_task_key_union
    ):

        continue


    counter = Counter(
        stable_json_key(
            row.get(
                field_name
            )
        )

        for row
        in cell54_legacy_task_rows
    )


    cell54_semantic_field_distributions[
        field_name
    ] = counter


# ============================================================
# 8) Compare ready_case_plans with task_draft_plan_rows
#
# For these 38 ready cases we expect semantic equivalence.
# ============================================================

cell54_task_row_by_case_id = {
    row[
        "golden_case_id"
    ]:
        row

    for row
    in cell54_legacy_task_rows
}


cell54_ready_row_by_case_id = {
    row[
        "golden_case_id"
    ]:
        row

    for row
    in cell54_legacy_ready_case_plan_rows
}


cell54_ready_vs_task_mismatches = []


for legacy_case_id in sorted(
    cell46_legacy_exercise_case_ids
):

    task_row = (
        cell54_task_row_by_case_id.get(
            legacy_case_id
        )
    )


    ready_row = (
        cell54_ready_row_by_case_id.get(
            legacy_case_id
        )
    )


    if (
        task_row
        is None

        or

        ready_row
        is None
    ):

        cell54_ready_vs_task_mismatches.append(
            {
                "golden_case_id":
                    legacy_case_id,

                "reason":
                    "MISSING_ROW",
            }
        )

        continue


    if (
        stable_json_key(
            task_row
        )
        !=
        stable_json_key(
            ready_row
        )
    ):

        cell54_ready_vs_task_mismatches.append(
            {
                "golden_case_id":
                    legacy_case_id,

                "reason":
                    "ROW_CONTENT_MISMATCH",
            }
        )


# ============================================================
# 9) Validate task_draft_plans index
# ============================================================

cell54_task_index_failures = []


for task_row in (
    cell54_legacy_task_rows
):

    task_draft_id = (
        task_row[
            "task_draft_id"
        ]
    )


    if (
        task_draft_id
        not in
        task_draft_plans
    ):

        cell54_task_index_failures.append(
            {
                "task_draft_id":
                    task_draft_id,

                "reason":
                    "MISSING_FROM_TASK_DRAFT_PLANS_INDEX",
            }
        )

        continue


    indexed_row = (
        task_draft_plans[
            task_draft_id
        ]
    )


    if (
        stable_json_key(
            indexed_row
        )
        !=
        stable_json_key(
            task_row
        )
    ):

        cell54_task_index_failures.append(
            {
                "task_draft_id":
                    task_draft_id,

                "reason":
                    "INDEXED_RECORD_MISMATCH",
            }
        )


# ============================================================
# 10) Search actual notebook source for Task-ID construction
# ============================================================

CELL54_SOURCE_TERMS = [
    "task_draft_id",
    "def build_task",
    "build_task_draft",
    "task-draft:",
    "question_archetype",
    "evidence_target_method",
]


cell54_candidate_source_cells = []


for code_cell in (
    cell49_code_cells
):

    source = (
        code_cell[
            "source"
        ]
    )


    matched_terms = [
        term

        for term
        in CELL54_SOURCE_TERMS

        if (
            term
            in
            source
        )
    ]


    if not (
        "task_draft_id"
        in
        matched_terms

        or

        "task-draft:"
        in
        matched_terms

        or

        "build_task_draft"
        in
        matched_terms
    ):

        continue


    cell54_candidate_source_cells.append(
        {
            "cell_index":
                code_cell[
                    "cell_index"
                ],

            "matched_terms":
                matched_terms,

            "source":
                source,
        }
    )


# ============================================================
# 11) Extract narrow source snippets
# ============================================================

cell54_formula_evidence_snippets = []


for source_row in (
    cell54_candidate_source_cells
):

    source_lines = (
        source_row[
            "source"
        ].splitlines()
    )


    matched_indexes = [
        index

        for index, line
        in enumerate(
            source_lines
        )

        if (
            "task_draft_id"
            in
            line

            or

            "task-draft:"
            in
            line

            or

            "def build_task"
            in
            line
        )
    ]


    for matched_index in (
        matched_indexes
    ):

        start_index = max(
            0,
            matched_index
            - 8,
        )


        end_index = min(
            len(
                source_lines
            ),
            matched_index
            + 20,
        )


        cell54_formula_evidence_snippets.append(
            {
                "cell_index":
                    source_row[
                        "cell_index"
                    ],

                "matched_line_index":
                    matched_index,

                "snippet":
                    "\n".join(
                        source_lines[
                            start_index:
                            end_index
                        ]
                    ),
            }
        )


# ============================================================
# 12) Discover Task-ID builder callables from fixed namespace
# ============================================================

cell54_candidate_task_id_callables = []


for variable_name, value in sorted(
    cell54_globals_snapshot_items,
    key=lambda item:
        item[
            0
        ],
):

    if not callable(
        value
    ):

        continue


    lowered_name = (
        variable_name.lower()
    )


    if not (
        "task"
        in
        lowered_name

        and

        "id"
        in
        lowered_name
    ):

        continue


    code_object = getattr(
        value,
        "__code__",
        None,
    )


    if (
        code_object
        is None
    ):

        continue


    cell54_candidate_task_id_callables.append(
        {
            "name":
                variable_name,

            "callable":
                value,

            "arg_names":
                list(
                    code_object.co_varnames[
                        :code_object.co_argcount
                    ]
                ),

            "referenced_names":
                list(
                    code_object.co_names
                ),

            "constants":
                [
                    constant

                    for constant
                    in code_object.co_consts

                    if isinstance(
                        constant,
                        (
                            str,
                            int,
                            float,
                            bool,
                            type(
                                None
                            ),
                        ),
                    )
                ],
        }
    )


# ============================================================
# 13) Build per-case validation context
#
# Combines existing case, task, design and precheck records.
# No value is invented.
# ============================================================

cell54_validation_context_by_case_id = {}


for task_row in (
    cell54_legacy_task_rows
):

    legacy_case_id = (
        task_row[
            "golden_case_id"
        ]
    )


    merged_context = {}


    merged_context.update(
        golden_case_records_by_id[
            legacy_case_id
        ]
    )


    merged_context.update(
        task_row
    )


    groundability_record = (
        cell54_resolve_case_registry_record(
            case_groundability_precheck,
            legacy_case_id,
        )
    )


    task_design_record = (
        cell54_resolve_case_registry_record(
            case_task_design,
            legacy_case_id,
        )
    )


    if (
        groundability_record
        is not None
    ):

        merged_context.update(
            groundability_record
        )


    if (
        task_design_record
        is not None
    ):

        merged_context.update(
            task_design_record
        )


    cell54_validation_context_by_case_id[
        legacy_case_id
    ] = merged_context


# ============================================================
# 14) Safe behavioral validation of discovered builders
#
# Only execute a candidate if every required positional arg
# exists in every merged legacy validation context.
# ============================================================

cell54_task_builder_validation_rows = []


for candidate in (
    cell54_candidate_task_id_callables
):

    arg_names = (
        candidate[
            "arg_names"
        ]
    )


    all_args_available = all(
        all(
            arg_name
            in
            cell54_validation_context_by_case_id[
                legacy_case_id
            ]

            for arg_name
            in arg_names
        )

        for legacy_case_id
        in cell46_legacy_exercise_case_ids
    )


    if not (
        all_args_available
    ):

        cell54_task_builder_validation_rows.append(
            {
                "builder_name":
                    candidate[
                        "name"
                    ],

                "arg_names":
                    arg_names,

                "executed":
                    False,

                "exact_match_count":
                    0,

                "error_count":
                    0,

                "reproduces_all_38":
                    False,

                "skip_reason":
                    "REQUIRED_ARGUMENT_NOT_AVAILABLE_IN_LEGACY_CONTEXT",
            }
        )

        continue


    exact_matches = 0

    errors = []

    mismatches = []


    for task_row in (
        cell54_legacy_task_rows
    ):

        legacy_case_id = (
            task_row[
                "golden_case_id"
            ]
        )


        context = (
            cell54_validation_context_by_case_id[
                legacy_case_id
            ]
        )


        kwargs = {
            arg_name:
                context[
                    arg_name
                ]

            for arg_name
            in arg_names
        }


        try:

            produced_task_id = (
                candidate[
                    "callable"
                ](
                    **kwargs
                )
            )


        except Exception as exc:

            errors.append(
                {
                    "golden_case_id":
                        legacy_case_id,

                    "error_type":
                        type(
                            exc
                        ).__name__,

                    "error_message":
                        str(
                            exc
                        ),
                }
            )

            continue


        if (
            produced_task_id
            ==
            task_row[
                "task_draft_id"
            ]
        ):

            exact_matches += 1


        else:

            mismatches.append(
                {
                    "golden_case_id":
                        legacy_case_id,

                    "stored_task_draft_id":
                        task_row[
                            "task_draft_id"
                        ],

                    "produced_task_draft_id":
                        produced_task_id,
                }
            )


    cell54_task_builder_validation_rows.append(
        {
            "builder_name":
                candidate[
                    "name"
            ],

            "arg_names":
                arg_names,

            "executed":
                True,

            "exact_match_count":
                exact_matches,

            "mismatch_count":
                len(
                    mismatches
                ),

            "error_count":
                len(
                    errors
                ),

            "errors":
                errors,

            "mismatches":
                mismatches,

            "reproduces_all_38":
                (
                    exact_matches
                    == 38

                    and

                    len(
                        errors
                    )
                    == 0

                    and

                    len(
                        mismatches
                    )
                    == 0
                ),
        }
    )


cell54_exact_task_id_builders = [
    row

    for row
    in cell54_task_builder_validation_rows

    if (
        row[
            "reproduces_all_38"
        ]
        is True
    )
]


# ============================================================
# 15) Resolve Task-ID formula status
# ============================================================

if (
    len(
        cell54_exact_task_id_builders
    )
    == 1
):

    cell54_task_id_formula_status = (
        "RESOLVED_BY_EXISTING_BUILDER_38_OF_38"
    )


    cell54_resolved_task_id_builder_name = (
        cell54_exact_task_id_builders[
            0
        ][
            "builder_name"
        ]
    )


else:

    cell54_task_id_formula_status = (
        "NOT_YET_RESOLVED"
    )


    cell54_resolved_task_id_builder_name = None


# ============================================================
# 16) Independent second-pass reproduction
# ============================================================

cell54_independent_task_id_failures = []


if (
    cell54_resolved_task_id_builder_name
    is not None
):

    cell54_resolved_task_builder = (
        cell54_globals_snapshot[
            cell54_resolved_task_id_builder_name
        ]
    )


    cell54_builder_code = (
        cell54_resolved_task_builder.__code__
    )


    cell54_builder_arg_names = list(
        cell54_builder_code.co_varnames[
            :cell54_builder_code.co_argcount
        ]
    )


    for task_row in (
        cell54_legacy_task_rows
    ):

        legacy_case_id = (
            task_row[
                "golden_case_id"
            ]
        )


        context = (
            cell54_validation_context_by_case_id[
                legacy_case_id
            ]
        )


        kwargs = {
            arg_name:
                context[
                    arg_name
                ]

            for arg_name
            in cell54_builder_arg_names
        }


        independently_recomputed_id = (
            cell54_resolved_task_builder(
                **kwargs
            )
        )


        if (
            independently_recomputed_id
            !=
            task_row[
                "task_draft_id"
            ]
        ):

            cell54_independent_task_id_failures.append(
                {
                    "golden_case_id":
                        legacy_case_id,

                    "stored":
                        task_row[
                            "task_draft_id"
                        ],

                    "recomputed":
                        independently_recomputed_id,
                }
            )


# ============================================================
# 17) Identify fields needing explicit V2 decision
#
# This is a classification only.
# ============================================================

CELL54_ALWAYS_RECOMPUTE_FOR_V2 = {
    "golden_case_id",
    "task_draft_id",
}


CELL54_EVIDENCE_SEMANTIC_FIELDS = {
    "evidence_scope",
    "evidence_target_method",
    "target_evidence_status",
    "citation_target_status",
}


cell54_v2_task_field_policy = {}


for field_name in sorted(
    cell54_task_key_union
):

    if (
        field_name
        in
        CELL54_ALWAYS_RECOMPUTE_FOR_V2
    ):

        policy = (
            "RECOMPUTE_FOR_V2_CASE"
        )


    elif (
        field_name
        in
        CELL54_EVIDENCE_SEMANTIC_FIELDS
    ):

        policy = (
            "EXPLICIT_V2_EVIDENCE_METHOD_DECISION_REQUIRED"
        )


    elif (
        field_name
        in
        cell54_task_constant_fields
    ):

        policy = (
            "CANDIDATE_FOR_PRESERVATION_AFTER_SEMANTIC_REVIEW"
        )


    else:

        policy = (
            "REBUILD_FROM_V2_CASE_OR_TASK_DESIGN"
        )


    cell54_v2_task_field_policy[
        field_name
    ] = policy


cell54_fields_requiring_explicit_v2_evidence_decision = sorted(
    [
        field_name

        for field_name, policy
        in cell54_v2_task_field_policy.items()

        if (
            policy
            ==
            "EXPLICIT_V2_EVIDENCE_METHOD_DECISION_REQUIRED"
        )
    ]
)


# ============================================================
# 18) Task Contract Discovery registry
# ============================================================

EXERCISE_V2_TASK_CONTRACT_DISCOVERY_V1_DRAFT = {
    "version":
        "exercise_v2_task_contract_discovery_v1_draft",

    "status":
        (
            "LEGACY_TASK_CONTRACT_AND_ID_FORMULA_RESOLVED"
            if (
                cell54_task_id_formula_status
                ==
                "RESOLVED_BY_EXISTING_BUILDER_38_OF_38"
            )
            else
            "TASK_ID_FORMULA_NOT_YET_RESOLVED"
        ),

    "legacy_exercise_task_count":
        len(
            cell54_legacy_task_rows
        ),

    "legacy_task_schema_uniform":
        cell54_task_schema_uniform,

    "task_id_formula_status":
        cell54_task_id_formula_status,

    "resolved_task_id_builder_name":
        cell54_resolved_task_id_builder_name,

    "v2_evidence_decision_fields":
        list(
            cell54_fields_requiring_explicit_v2_evidence_decision
        ),

    "v2_task_staging_status":
        "NOT_STARTED",

    "case_activation_status":
        "NOT_STARTED",

    "target_rebuild_status":
        "NOT_STARTED",

    "pilot_rebuild_status":
        "NOT_STARTED",

    "question_generation_status":
        "NOT_STARTED",

    "artifact_write_status":
        "NONE",
}


# ============================================================
# 19) No-mutation recheck
# ============================================================

cell54_task_snapshot_after = hashlib.sha256(
    stable_json_key(
        task_draft_plan_rows
    ).encode(
        "utf-8"
    )
).hexdigest()


cell54_ready_plan_snapshot_after = hashlib.sha256(
    stable_json_key(
        ready_case_plans
    ).encode(
        "utf-8"
    )
).hexdigest()


cell54_design_snapshot_after = hashlib.sha256(
    stable_json_key(
        case_task_design
    ).encode(
        "utf-8"
    )
).hexdigest()


cell54_groundability_snapshot_after = hashlib.sha256(
    stable_json_key(
        case_groundability_precheck
    ).encode(
        "utf-8"
    )
).hexdigest()


cell54_target_snapshot_after = hashlib.sha256(
    stable_json_key(
        target_evidence_records
    ).encode(
        "utf-8"
    )
).hexdigest()


cell54_pilot_snapshot_after = hashlib.sha256(
    stable_json_key(
        pilot_selection_rows
    ).encode(
        "utf-8"
    )
).hexdigest()


cell54_golden_snapshot_after = hashlib.sha256(
    stable_json_key(
        golden_case_records
    ).encode(
        "utf-8"
    )
).hexdigest()


cell54_staged_case_snapshot_after = hashlib.sha256(
    stable_json_key(
        staged_exercise_v2_golden_case_records
    ).encode(
        "utf-8"
    )
).hexdigest()


# ============================================================
# 20) Validation gates
# ============================================================

exact_38_task_rows_gate = (
    len(
        cell54_legacy_task_rows
    )
    == 38
)


exact_38_ready_rows_gate = (
    len(
        cell54_legacy_ready_case_plan_rows
    )
    == 38
)


uniform_task_schema_gate = (
    cell54_task_schema_uniform

    and

    cell54_task_key_union
    ==
    cell54_task_key_intersection
)


exact_38_groundability_records_gate = (
    len(
        cell54_legacy_groundability_rows
    )
    == 38

    and

    len(
        cell54_missing_groundability_cases
    )
    == 0
)


exact_38_task_design_records_gate = (
    len(
        cell54_legacy_task_design_rows
    )
    == 38

    and

    len(
        cell54_missing_task_design_cases
    )
    == 0
)


ready_and_task_rows_match_gate = (
    len(
        cell54_ready_vs_task_mismatches
    )
    == 0
)


task_index_exact_gate = (
    len(
        cell54_task_index_failures
    )
    == 0
)


source_formula_evidence_found_gate = (
    len(
        cell54_candidate_source_cells
    )
    > 0

    and

    len(
        cell54_formula_evidence_snippets
    )
    > 0
)


exactly_one_task_id_builder_gate = (
    len(
        cell54_exact_task_id_builders
    )
    == 1
)


task_id_formula_resolved_gate = (
    cell54_task_id_formula_status
    ==
    "RESOLVED_BY_EXISTING_BUILDER_38_OF_38"
)


independent_task_id_reproduction_gate = (
    task_id_formula_resolved_gate

    and

    len(
        cell54_independent_task_id_failures
    )
    == 0
)


all_task_fields_classified_gate = (
    set(
        cell54_v2_task_field_policy.keys()
    )
    ==
    cell54_task_key_union
)


v2_task_staging_not_started_gate = (
    EXERCISE_V2_TASK_CONTRACT_DISCOVERY_V1_DRAFT[
        "v2_task_staging_status"
    ]
    ==
    "NOT_STARTED"
)


active_registries_unchanged_gate = (
    cell54_task_snapshot_before
    ==
    cell54_task_snapshot_after

    and

    cell54_ready_plan_snapshot_before
    ==
    cell54_ready_plan_snapshot_after

    and

    cell54_design_snapshot_before
    ==
    cell54_design_snapshot_after

    and

    cell54_groundability_snapshot_before
    ==
    cell54_groundability_snapshot_after

    and

    cell54_target_snapshot_before
    ==
    cell54_target_snapshot_after

    and

    cell54_pilot_snapshot_before
    ==
    cell54_pilot_snapshot_after

    and

    cell54_golden_snapshot_before
    ==
    cell54_golden_snapshot_after

    and

    cell54_staged_case_snapshot_before
    ==
    cell54_staged_case_snapshot_after

    and

    len(
        golden_case_records
    )
    == 724

    and

    len(
        task_draft_plan_rows
    )
    == 724

    and

    len(
        target_evidence_records
    )
    == 572

    and

    len(
        pilot_selection_rows
    )
    == 23
)


case_activation_not_started_gate = (
    EXERCISE_V2_TASK_CONTRACT_DISCOVERY_V1_DRAFT[
        "case_activation_status"
    ]
    ==
    "NOT_STARTED"
)


target_rebuild_not_started_gate = (
    EXERCISE_V2_TASK_CONTRACT_DISCOVERY_V1_DRAFT[
        "target_rebuild_status"
    ]
    ==
    "NOT_STARTED"
)


pilot_rebuild_not_started_gate = (
    EXERCISE_V2_TASK_CONTRACT_DISCOVERY_V1_DRAFT[
        "pilot_rebuild_status"
    ]
    ==
    "NOT_STARTED"
)


question_generation_not_started_gate = (
    EXERCISE_V2_TASK_CONTRACT_DISCOVERY_V1_DRAFT[
        "question_generation_status"
    ]
    ==
    "NOT_STARTED"
)


no_artifact_write_gate = (
    EXERCISE_V2_TASK_CONTRACT_DISCOVERY_V1_DRAFT[
        "artifact_write_status"
    ]
    ==
    "NONE"
)


cell_54_gates = {
    "exact_38_legacy_exercise_task_draft_rows":
        exact_38_task_rows_gate,

    "exact_38_legacy_ready_case_plan_rows":
        exact_38_ready_rows_gate,

    "legacy_exercise_task_schema_uniform":
        uniform_task_schema_gate,

    "exact_38_groundability_records_resolved":
        exact_38_groundability_records_gate,

    "exact_38_case_task_design_records_resolved":
        exact_38_task_design_records_gate,

    "ready_case_plans_and_task_draft_rows_match_for_all_38":
        ready_and_task_rows_match_gate,

    "all_38_task_draft_ids_resolve_exactly_in_task_index":
        task_index_exact_gate,

    "actual_notebook_source_contains_task_id_formula_evidence":
        source_formula_evidence_found_gate,

    "exactly_one_existing_task_id_builder_reproduces_all_38":
        exactly_one_task_id_builder_gate,

    "task_draft_id_formula_resolved_without_guessing":
        task_id_formula_resolved_gate,

    "resolved_task_builder_independently_reproduces_38_of_38":
        independent_task_id_reproduction_gate,

    "every_task_field_has_explicit_v2_handling_policy":
        all_task_fields_classified_gate,

    "v2_task_staging_not_started":
        v2_task_staging_not_started_gate,

    "all_active_and_case_staging_registries_content_unchanged":
        active_registries_unchanged_gate,

    "case_activation_not_started":
        case_activation_not_started_gate,

    "target_rebuild_not_started":
        target_rebuild_not_started_gate,

    "pilot_rebuild_not_started":
        pilot_rebuild_not_started_gate,

    "question_generation_not_started":
        question_generation_not_started_gate,

    "no_artifact_write":
        no_artifact_write_gate,
}


cell_54_pass = all(
    cell_54_gates.values()
)


# ============================================================
# 21) Output
# ============================================================

print(
    "=== LEGACY EXERCISE TASK/PLAN CONTRACT DISCOVERY ==="
)


print(
    "Legacy Task Draft rows:",
    len(
        cell54_legacy_task_rows
    )
)


print(
    "Legacy Ready Case Plan rows:",
    len(
        cell54_legacy_ready_case_plan_rows
    )
)


print(
    "Uniform Task schema:",
    cell54_task_schema_uniform
)


print(
    "Task fields:",
    sorted(
        cell54_task_key_union
    )
)


print(
    "\n=== TASK FIELD DISTINCT COUNTS ==="
)


for field_name, count in sorted(
    cell54_task_field_distinct_counts.items()
):

    print(
        "-",
        field_name,
        ":",
        count,
    )


print(
    "\n=== CONSTANT TASK FIELDS ==="
)


for field_name, value in sorted(
    cell54_task_constant_fields.items()
):

    print(
        "-",
        field_name,
        "=",
        value,
    )


print(
    "\n=== SEMANTIC TASK FIELD DISTRIBUTIONS ==="
)


for field_name, counter in (
    cell54_semantic_field_distributions.items()
):

    print(
        "\n",
        field_name,
        sep="",
    )


    for serialized_value, count in (
        counter.items()
    ):

        print(
            "- count:",
            count,
            "| value:",
            serialized_value,
        )


print(
    "\n=== TASK REGISTRY CONSISTENCY ==="
)


print(
    "Missing Groundability records:",
    len(
        cell54_missing_groundability_cases
    )
)


print(
    "Missing Task Design records:",
    len(
        cell54_missing_task_design_cases
    )
)


print(
    "Ready/Task mismatches:",
    len(
        cell54_ready_vs_task_mismatches
    )
)


print(
    "Task index failures:",
    len(
        cell54_task_index_failures
    )
)


print(
    "\n=== TASK-ID BUILDER CANDIDATES ==="
)


if (
    cell54_candidate_task_id_callables
):

    for row in (
        cell54_candidate_task_id_callables
    ):

        print(
            "-",
            row[
                "name"
            ],
            "| args:",
            row[
                "arg_names"
            ],
            "| referenced:",
            row[
                "referenced_names"
            ],
        )

else:

    print(
        "NONE"
    )


print(
    "\n=== TASK-ID BUILDER VALIDATION ==="
)


for row in (
    cell54_task_builder_validation_rows
):

    print(
        "-",
        row[
            "builder_name"
        ],
        "| executed:",
        row[
            "executed"
        ],
        "| matches:",
        row[
            "exact_match_count"
        ],
        "/ 38",
        "| errors:",
        row[
            "error_count"
        ],
        "| exact:",
        row[
            "reproduces_all_38"
        ],
    )


print(
    "\n=== TASK-ID FORMULA RESOLUTION ==="
)


print(
    "Status:",
    cell54_task_id_formula_status
)


print(
    "Resolved builder:",
    cell54_resolved_task_id_builder_name
)


print(
    "Independent reproduction failures:",
    len(
        cell54_independent_task_id_failures
    )
)


print(
    "\n=== SOURCE FORMULA EVIDENCE ==="
)


for index, row in enumerate(
    cell54_formula_evidence_snippets[
        :12
    ],
    start=1,
):

    print(
        "\nSNIPPET",
        index,
        "| notebook cell:",
        row[
            "cell_index"
        ],
        "| line:",
        row[
            "matched_line_index"
        ],
    )

    print(
        row[
            "snippet"
        ]
    )


print(
    "\n=== V2 TASK FIELD POLICY ==="
)


for field_name, policy in sorted(
    cell54_v2_task_field_policy.items()
):

    print(
        "-",
        field_name,
        "=>",
        policy,
    )


print(
    "\nFields requiring explicit V2 evidence-method decision:",
    cell54_fields_requiring_explicit_v2_evidence_decision
)


print(
    "\n=== ACTIVE COUNTS ==="
)


print(
    "Golden:",
    len(
        golden_case_records
    )
)


print(
    "Task Plans:",
    len(
        task_draft_plan_rows
    )
)


print(
    "Targets:",
    len(
        target_evidence_records
    )
)


print(
    "Pilots:",
    len(
        pilot_selection_rows
    )
)


print(
    "\n=== CODE CELL 54 GATES ==="
)


for name, passed in (
    cell_54_gates.items()
):

    print(
        name,
        ":",
        "PASS"
        if passed
        else "FAIL"
    )


print(
    "\nCODE CELL 54 GATE:",
    "PASS"
    if cell_54_pass
    else "FAIL"
)


assert cell_54_pass, (
    "Exercise Task/Plan contract discovery failed. "
    "Do not stage V2 Tasks yet. Repair this SAME Code Cell 54 "
    "until the actual Task Draft ID formula, all 38 Task/Design/"
    "Groundability dependencies and V2 field-handling policy "
    "are fully resolved without mutating active registries."
)
```

```python
# Code Cell 56

# ============================================================
# Exercise V2 Groundability + Task Design + Task Draft Staging
#
# Staging only.
#
# Builds:
#   1) staged_exercise_v2_groundability_precheck
#   2) staged_exercise_v2_case_task_design
#   3) staged_exercise_v2_task_draft_plan_rows
#
# Does NOT mutate:
#   case_groundability_precheck
#   case_task_design
#   task_draft_plans
#   task_draft_plan_rows
#   ready_case_plans
#   target_evidence_records
#   pilot_selection_rows
#   golden_case_records
#
# NO Case activation.
# NO Target rebuild.
# NO Pilot rebuild.
# NO question generation.
# NO artifact writes.
# ============================================================


# ============================================================
# 1) Active-state snapshots
# ============================================================

cell56_groundability_snapshot_before = hashlib.sha256(
    stable_json_key(
        case_groundability_precheck
    ).encode(
        "utf-8"
    )
).hexdigest()


cell56_design_snapshot_before = hashlib.sha256(
    stable_json_key(
        case_task_design
    ).encode(
        "utf-8"
    )
).hexdigest()


cell56_task_rows_snapshot_before = hashlib.sha256(
    stable_json_key(
        task_draft_plan_rows
    ).encode(
        "utf-8"
    )
).hexdigest()


cell56_task_index_snapshot_before = hashlib.sha256(
    stable_json_key(
        task_draft_plans
    ).encode(
        "utf-8"
    )
).hexdigest()


cell56_ready_snapshot_before = hashlib.sha256(
    stable_json_key(
        ready_case_plans
    ).encode(
        "utf-8"
    )
).hexdigest()


cell56_target_snapshot_before = hashlib.sha256(
    stable_json_key(
        target_evidence_records
    ).encode(
        "utf-8"
    )
).hexdigest()


cell56_pilot_snapshot_before = hashlib.sha256(
    stable_json_key(
        pilot_selection_rows
    ).encode(
        "utf-8"
    )
).hexdigest()


cell56_golden_snapshot_before = hashlib.sha256(
    stable_json_key(
        golden_case_records
    ).encode(
        "utf-8"
    )
).hexdigest()


cell56_case_staging_snapshot_before = hashlib.sha256(
    stable_json_key(
        staged_exercise_v2_golden_case_records
    ).encode(
        "utf-8"
    )
).hexdigest()


# ============================================================
# 2) Audit helper for legacy dependency families
# ============================================================

def cell56_audit_record_family(
    rows,
):

    if not (
        rows
    ):

        return {
            "row_count":
                0,

            "schema_uniform":
                False,

            "key_union":
                set(),

            "key_intersection":
                set(),

            "distinct_count_by_field":
                {},

            "constant_value_by_field":
                {},

            "varying_fields":
                [],
        }


    key_sets = [
        set(
            row.keys()
        )
        for row
        in rows
    ]


    key_union = set().union(
        *key_sets
    )


    key_intersection = set(
        key_sets[
            0
        ]
    )


    for key_set in (
        key_sets[
            1:
        ]
    ):

        key_intersection &= (
            key_set
        )


    distinct_count_by_field = {}

    constant_value_by_field = {}


    for field_name in sorted(
        key_union
    ):

        values = [
            row.get(
                field_name
            )

            for row
            in rows
        ]


        distinct_serialized = {
            stable_json_key(
                value
            )
            for value
            in values
        }


        distinct_count_by_field[
            field_name
        ] = len(
            distinct_serialized
        )


        if (
            len(
                distinct_serialized
            )
            == 1
        ):

            constant_value_by_field[
                field_name
            ] = values[
                0
            ]


    varying_fields = sorted(
        [
            field_name

            for field_name, count
            in distinct_count_by_field.items()

            if (
                count
                > 1
            )
        ]
    )


    return {
        "row_count":
            len(
                rows
            ),

        "schema_uniform":
            (
                key_union
                ==
                key_intersection
            ),

        "key_union":
            key_union,

        "key_intersection":
            key_intersection,

        "distinct_count_by_field":
            distinct_count_by_field,

        "constant_value_by_field":
            constant_value_by_field,

        "varying_fields":
            varying_fields,
    }


# ============================================================
# 3) Exact Legacy dependency audits
# ============================================================

cell56_groundability_audit = (
    cell56_audit_record_family(
        cell54_legacy_groundability_rows
    )
)


cell56_task_design_audit = (
    cell56_audit_record_family(
        cell54_legacy_task_design_rows
    )
)


print(
    "=== LEGACY DEPENDENCY SCHEMA AUDIT ==="
)


print(
    "Groundability rows:",
    cell56_groundability_audit[
        "row_count"
    ]
)


print(
    "Groundability schema uniform:",
    cell56_groundability_audit[
        "schema_uniform"
    ]
)


print(
    "Groundability fields:",
    sorted(
        cell56_groundability_audit[
            "key_union"
        ]
    )
)


print(
    "Groundability varying fields:",
    cell56_groundability_audit[
        "varying_fields"
    ]
)


print(
    "\nTask Design rows:",
    cell56_task_design_audit[
        "row_count"
    ]
)


print(
    "Task Design schema uniform:",
    cell56_task_design_audit[
        "schema_uniform"
    ]
)


print(
    "Task Design fields:",
    sorted(
        cell56_task_design_audit[
            "key_union"
        ]
    )
)


print(
    "Task Design varying fields:",
    cell56_task_design_audit[
        "varying_fields"
    ]
)


# ============================================================
# 4) Explicit V2 derivation policy
#
# If any of these fields exist in a Legacy dependency record,
# their V2 values MUST be derived from the staged V2 Case /
# semantic contract rather than copied.
# ============================================================

CELL56_EXPLICIT_DERIVABLE_FIELDS = {
    "golden_case_id",
    "case_schema_version",
    "source_type",
    "stratum",
    "candidate_identity",
    "candidate_identity_sha256",
    "evidence_unit",
    "evidence_document_ids",
    "evidence_document_count",
    "membership_status",
    "final_golden_membership",
    "review_status",
    "task_family",
    "question_archetype",
    "evidence_scope",
    "evidence_target_method",
    "required_dimensions",
    "authority_expectation",
    "context_expectation",
    "construction_eligibility",
    "target_evidence_status",
    "question_text_status",
    "expected_answer_status",
    "citation_target_status",
}


CELL56_FORBIDDEN_STALE_VALUES = {
    "BOOK_REGION",
    "USE_EXPLICIT_BOOK_REGION",
    "SELECTED_AND_VALIDATED",
    "READY_FOR_QUESTION_DRAFT",
    "BLOCKED_PENDING_HUMAN_CONFIRMATION",
}


# ============================================================
# 5) Determine whether any varying Legacy field lacks a V2
# derivation rule.
# ============================================================

cell56_unresolved_groundability_varying_fields = sorted(
    set(
        cell56_groundability_audit[
            "varying_fields"
        ]
    )
    -
    CELL56_EXPLICIT_DERIVABLE_FIELDS
)


cell56_unresolved_design_varying_fields = sorted(
    set(
        cell56_task_design_audit[
            "varying_fields"
        ]
    )
    -
    CELL56_EXPLICIT_DERIVABLE_FIELDS
)


# ============================================================
# 6) Detect stale constant semantics that would otherwise be
# silently preserved.
# ============================================================

def cell56_detect_unmapped_stale_constants(
    audit,
):

    failures = []


    for field_name, value in (
        audit[
            "constant_value_by_field"
        ].items()
    ):

        if (
            field_name
            in
            CELL56_EXPLICIT_DERIVABLE_FIELDS
        ):

            continue


        if (
            isinstance(
                value,
                str,
            )

            and

            value
            in
            CELL56_FORBIDDEN_STALE_VALUES
        ):

            failures.append(
                {
                    "field":
                        field_name,

                    "value":
                        value,
                }
            )


    return failures


cell56_groundability_stale_constant_failures = (
    cell56_detect_unmapped_stale_constants(
        cell56_groundability_audit
    )
)


cell56_design_stale_constant_failures = (
    cell56_detect_unmapped_stale_constants(
        cell56_task_design_audit
    )
)


# ============================================================
# 7) V2 Case lookup
# ============================================================

cell56_v2_case_by_id = {
    record[
        "golden_case_id"
    ]:
        record

    for record
    in staged_exercise_v2_golden_case_records
}


cell56_v2_case_by_population_unit = {
    record[
        "candidate_identity"
    ][
        "population_unit_id"
    ]:
        record

    for record
    in staged_exercise_v2_golden_case_records
}


assert (
    len(
        cell56_v2_case_by_id
    )
    ==
    38
)


assert (
    len(
        cell56_v2_case_by_population_unit
    )
    ==
    38
)


# ============================================================
# 8) Build all explicit per-case V2 derivation values
# ============================================================

def cell56_explicit_v2_values(
    staged_case,
):

    return {
        "golden_case_id":
            staged_case[
                "golden_case_id"
            ],

        "case_schema_version":
            staged_case[
                "case_schema_version"
            ],

        "source_type":
            staged_case[
                "source_type"
            ],

        "stratum":
            staged_case[
                "stratum"
            ],

        "candidate_identity":
            cell52_safe_clone(
                staged_case[
                    "candidate_identity"
                ]
            ),

        "candidate_identity_sha256":
            staged_case[
                "candidate_identity_sha256"
            ],

        "evidence_unit":
            staged_case[
                "evidence_unit"
            ],

        "evidence_document_ids":
            list(
                staged_case[
                    "evidence_document_ids"
                ]
            ),

        "evidence_document_count":
            staged_case[
                "evidence_document_count"
            ],

        "membership_status":
            staged_case[
                "membership_status"
            ],

        "final_golden_membership":
            staged_case[
                "final_golden_membership"
            ],

        "review_status":
            staged_case[
                "review_status"
            ],

        **cell52_safe_clone(
            EXERCISE_V2_PRETARGET_TASK_SEMANTIC_TEMPLATE
        ),
    }


# ============================================================
# 9) Generic safe transformer for Legacy dependency records
#
# - explicit V2 fields are replaced;
# - constant non-sensitive fields are preserved;
# - unresolved varying fields are forbidden.
# ============================================================

def cell56_transform_dependency_template(
    legacy_audit,
    staged_case,
):

    explicit_values = (
        cell56_explicit_v2_values(
            staged_case
        )
    )


    output = {}


    for field_name in sorted(
        legacy_audit[
            "key_union"
        ]
    ):

        if (
            field_name
            in
            explicit_values
        ):

            output[
                field_name
            ] = (
                cell52_safe_clone(
                    explicit_values[
                        field_name
                    ]
                )
            )

            continue


        if (
            field_name
            in
            legacy_audit[
                "constant_value_by_field"
            ]
        ):

            output[
                field_name
            ] = (
                cell52_safe_clone(
                    legacy_audit[
                        "constant_value_by_field"
                    ][
                        field_name
                    ]
                )
            )

            continue


        raise AssertionError(
            "Legacy dependency field varies across Exercise "
            "cases and has no explicit V2 derivation: "
            f"{field_name}"
        )


    return output


# ============================================================
# 10) Staging containers
# ============================================================

staged_exercise_v2_groundability_precheck = {}

staged_exercise_v2_case_task_design = {}

staged_exercise_v2_task_draft_plan_rows = []

staged_exercise_v2_task_draft_plans = {}


cell56_groundability_build_failures = []

cell56_task_design_build_failures = []

cell56_task_draft_build_failures = []


# ============================================================
# 11) Stage Groundability + Task Design records
#
# Only proceed when schemas and varying-field policy are safe.
# ============================================================

cell56_dependency_schema_safe_to_stage = (
    cell56_groundability_audit[
        "schema_uniform"
    ]

    and

    cell56_task_design_audit[
        "schema_uniform"
    ]

    and

    len(
        cell56_unresolved_groundability_varying_fields
    )
    == 0

    and

    len(
        cell56_unresolved_design_varying_fields
    )
    == 0

    and

    len(
        cell56_groundability_stale_constant_failures
    )
    == 0

    and

    len(
        cell56_design_stale_constant_failures
    )
    == 0
)


if (
    cell56_dependency_schema_safe_to_stage
):

    for staged_case in (
        staged_exercise_v2_golden_case_records
    ):

        v2_case_id = (
            staged_case[
                "golden_case_id"
            ]
        )


        # ----------------------------------------------------
        # Groundability
        # ----------------------------------------------------

        try:

            staged_groundability = (
                cell56_transform_dependency_template(
                    cell56_groundability_audit,
                    staged_case,
                )
            )


            if (
                v2_case_id
                in
                staged_exercise_v2_groundability_precheck
            ):

                raise AssertionError(
                    "Duplicate staged Groundability Case ID."
                )


            staged_exercise_v2_groundability_precheck[
                v2_case_id
            ] = (
                staged_groundability
            )


        except Exception as exc:

            cell56_groundability_build_failures.append(
                {
                    "golden_case_id":
                        v2_case_id,

                    "error_type":
                        type(
                            exc
                        ).__name__,

                    "error_message":
                        str(
                            exc
                        ),
                }
            )


        # ----------------------------------------------------
        # Case Task Design
        # ----------------------------------------------------

        try:

            staged_design = (
                cell56_transform_dependency_template(
                    cell56_task_design_audit,
                    staged_case,
                )
            )


            if (
                v2_case_id
                in
                staged_exercise_v2_case_task_design
            ):

                raise AssertionError(
                    "Duplicate staged Task Design Case ID."
                )


            staged_exercise_v2_case_task_design[
                v2_case_id
            ] = (
                staged_design
            )


        except Exception as exc:

            cell56_task_design_build_failures.append(
                {
                    "golden_case_id":
                        v2_case_id,

                    "error_type":
                        type(
                            exc
                        ).__name__,

                    "error_message":
                        str(
                            exc
                        ),
                }
            )


# ============================================================
# 12) Task Draft schema from actual Notebook
# ============================================================

cell56_task_schema_required_fields = list(
    TASK_DRAFT_SCHEMA_V1_DRAFT[
        "required_fields"
    ]
)


cell56_expected_task_fields = {
    "task_draft_id",
    "golden_case_id",
    "source_type",
    "stratum",
    "task_family",
    "question_archetype",
    "evidence_scope",
    "evidence_target_method",
    "required_dimensions",
    "authority_expectation",
    "context_expectation",
    "construction_eligibility",
    "target_evidence_status",
    "question_text_status",
    "expected_answer_status",
    "citation_target_status",
}


cell56_task_schema_exactly_expected = (
    set(
        cell56_task_schema_required_fields
    )
    ==
    cell56_expected_task_fields
)


# ============================================================
# 13) Stage exact V2 Task Draft records
# ============================================================

if (
    cell56_dependency_schema_safe_to_stage

    and

    cell56_task_schema_exactly_expected
):

    for staged_case in (
        staged_exercise_v2_golden_case_records
    ):

        v2_case_id = (
            staged_case[
                "golden_case_id"
            ]
        )


        try:

            semantic_template = (
                cell52_safe_clone(
                    EXERCISE_V2_PRETARGET_TASK_SEMANTIC_TEMPLATE
                )
            )


            task_draft_id = (
                build_task_draft_id(
                    golden_case_id=(
                        v2_case_id
                    ),
                    task_family=(
                        semantic_template[
                            "task_family"
                        ]
                    ),
                    question_archetype=(
                        semantic_template[
                            "question_archetype"
                        ]
                    ),
                )
            )


            staged_task = {
                "task_draft_id":
                    task_draft_id,

                "golden_case_id":
                    v2_case_id,

                **semantic_template,
            }


            if (
                set(
                    staged_task.keys()
                )
                !=
                cell56_expected_task_fields
            ):

                raise AssertionError(
                    "Staged V2 Task Draft field set does not "
                    "match TASK_DRAFT_SCHEMA_V1_DRAFT."
                )


            if (
                task_draft_id
                in
                staged_exercise_v2_task_draft_plans
            ):

                raise AssertionError(
                    "Duplicate staged V2 task_draft_id."
                )


            staged_exercise_v2_task_draft_plan_rows.append(
                staged_task
            )


            staged_exercise_v2_task_draft_plans[
                task_draft_id
            ] = (
                staged_task
            )


        except Exception as exc:

            cell56_task_draft_build_failures.append(
                {
                    "golden_case_id":
                        v2_case_id,

                    "error_type":
                        type(
                            exc
                        ).__name__,

                    "error_message":
                        str(
                            exc
                        ),
                }
            )


# Deterministic staging order.
staged_exercise_v2_task_draft_plan_rows = sorted(
    staged_exercise_v2_task_draft_plan_rows,
    key=lambda row:
        row[
            "golden_case_id"
        ],
)


# ============================================================
# 14) Staged dependency Case-ID coverage
# ============================================================

cell56_staged_case_ids = set(
    cell56_v2_case_by_id.keys()
)


cell56_staged_groundability_case_ids = set(
    staged_exercise_v2_groundability_precheck.keys()
)


cell56_staged_design_case_ids = set(
    staged_exercise_v2_case_task_design.keys()
)


cell56_staged_task_case_ids = {
    row[
        "golden_case_id"
    ]

    for row
    in staged_exercise_v2_task_draft_plan_rows
}


# ============================================================
# 15) Independent Task Draft ID recomputation
#
# Use direct formula rather than calling build_task_draft_id.
# ============================================================

cell56_task_id_recomputation_failures = []


for task_row in (
    staged_exercise_v2_task_draft_plan_rows
):

    payload = {
        "method_version":
            TASK_DRAFT_METHOD_VERSION,

        "golden_case_id":
            task_row[
                "golden_case_id"
            ],

        "task_family":
            task_row[
                "task_family"
            ],

        "question_archetype":
            task_row[
                "question_archetype"
            ],
    }


    digest = hashlib.sha256(
        stable_json_key(
            payload
        ).encode(
            "utf-8"
        )
    ).hexdigest()


    independent_task_id = (
        "task-draft:"
        +
        digest[
            :32
        ]
    )


    if (
        independent_task_id
        !=
        task_row[
            "task_draft_id"
        ]
    ):

        cell56_task_id_recomputation_failures.append(
            {
                "golden_case_id":
                    task_row[
                        "golden_case_id"
                    ],

                "stored":
                    task_row[
                        "task_draft_id"
                    ],

                "recomputed":
                    independent_task_id,
            }
        )


# ============================================================
# 16) Task-ID collision audit with active 724 Tasks
# ============================================================

cell56_active_task_ids = set(
    task_draft_plans.keys()
)


cell56_staged_task_ids = set(
    staged_exercise_v2_task_draft_plans.keys()
)


cell56_task_id_collisions_with_active = sorted(
    cell56_staged_task_ids
    &
    cell56_active_task_ids
)


# ============================================================
# 17) Semantic validation of all 38 staged Tasks
# ============================================================

cell56_task_semantic_failures = []


for task_row in (
    staged_exercise_v2_task_draft_plan_rows
):

    for field_name, expected_value in (
        EXERCISE_V2_PRETARGET_TASK_SEMANTIC_TEMPLATE.items()
    ):

        if (
            task_row[
                field_name
            ]
            !=
            expected_value
        ):

            cell56_task_semantic_failures.append(
                {
                    "golden_case_id":
                        task_row[
                            "golden_case_id"
                    ],

                    "field":
                        field_name,

                    "actual":
                        task_row[
                            field_name
                        ],

                    "expected":
                        expected_value,
                }
            )


# ============================================================
# 18) Groundability / Design stale-semantics audit
# ============================================================

cell56_dependency_stale_semantic_failures = []


for registry_name, registry in [
    (
        "groundability",
        staged_exercise_v2_groundability_precheck,
    ),
    (
        "task_design",
        staged_exercise_v2_case_task_design,
    ),
]:

    for case_id, record in (
        registry.items()
    ):

        for field_name, value in (
            record.items()
        ):

            if (
                isinstance(
                    value,
                    str,
                )

                and

                value
                in
                CELL56_FORBIDDEN_STALE_VALUES
            ):

                cell56_dependency_stale_semantic_failures.append(
                    {
                        "registry":
                            registry_name,

                        "golden_case_id":
                            case_id,

                        "field":
                            field_name,

                        "value":
                            value,
                    }
                )


# ============================================================
# 19) Confirm all staged dependencies reference V2 Case IDs
# ============================================================

cell56_dependency_case_linkage_failures = []


for registry_name, registry in [
    (
        "groundability",
        staged_exercise_v2_groundability_precheck,
    ),
    (
        "task_design",
        staged_exercise_v2_case_task_design,
    ),
]:

    for case_id, record in (
        registry.items()
    ):

        if (
            case_id
            not in
            cell56_staged_case_ids
        ):

            cell56_dependency_case_linkage_failures.append(
                {
                    "registry":
                        registry_name,

                    "case_id":
                        case_id,

                    "reason":
                        "INDEX_CASE_NOT_IN_STAGED_V2_CASES",
                }
            )


        if (
            "golden_case_id"
            in
            record

            and

            record[
                "golden_case_id"
            ]
            !=
            case_id
        ):

            cell56_dependency_case_linkage_failures.append(
                {
                    "registry":
                        registry_name,

                    "case_id":
                        case_id,

                    "reason":
                        "INTERNAL_GOLDEN_CASE_ID_MISMATCH",
                }
            )


for task_row in (
    staged_exercise_v2_task_draft_plan_rows
):

    if (
        task_row[
            "golden_case_id"
        ]
        not in
        cell56_staged_case_ids
    ):

        cell56_dependency_case_linkage_failures.append(
            {
                "registry":
                    "task_draft",

                "case_id":
                    task_row[
                        "golden_case_id"
                    ],

                "reason":
                    "TASK_CASE_NOT_IN_STAGED_V2_CASES",
            }
        )


# ============================================================
# 20) Correctly keep Ready Case staging EMPTY
#
# These V2 Tasks are blocked only because Targets are pending.
# They must not enter ready_case_plans yet.
# ============================================================

staged_exercise_v2_ready_case_plans = []


cell56_no_pretarget_ready_leakage = (
    len(
        staged_exercise_v2_ready_case_plans
    )
    == 0

    and

    all(
        row[
            "construction_eligibility"
        ]
        ==
        "BLOCKED_PENDING_V2_TARGET_REBUILD"

        for row
        in staged_exercise_v2_task_draft_plan_rows
    )
)


# ============================================================
# 21) Staging snapshot
# ============================================================

cell56_staging_snapshot_payload = {
    "groundability":
        staged_exercise_v2_groundability_precheck,

    "task_design":
        staged_exercise_v2_case_task_design,

    "task_drafts":
        staged_exercise_v2_task_draft_plan_rows,
}


cell56_staging_snapshot_sha256 = hashlib.sha256(
    stable_json_key(
        cell56_staging_snapshot_payload
    ).encode(
        "utf-8"
    )
).hexdigest()


# ============================================================
# 22) Staging registry
# ============================================================

EXERCISE_V2_TASK_DEPENDENCY_STAGING_V1_DRAFT = {
    "version":
        "exercise_v2_task_dependency_staging_v1_draft",

    "status":
        (
            "STAGED_NOT_ACTIVATED"
            if (
                len(
                    staged_exercise_v2_groundability_precheck
                )
                == 38

                and

                len(
                    staged_exercise_v2_case_task_design
                )
                == 38

                and

                len(
                    staged_exercise_v2_task_draft_plan_rows
                )
                == 38
            )
            else
            "STAGING_INCOMPLETE"
        ),

    "groundability_count":
        len(
            staged_exercise_v2_groundability_precheck
        ),

    "task_design_count":
        len(
            staged_exercise_v2_case_task_design
        ),

    "task_draft_count":
        len(
            staged_exercise_v2_task_draft_plan_rows
        ),

    "ready_case_plan_count":
        len(
            staged_exercise_v2_ready_case_plans
        ),

    "staging_snapshot_sha256":
        cell56_staging_snapshot_sha256,

    "target_rebuild_status":
        "NOT_STARTED",

    "case_activation_status":
        "NOT_STARTED",

    "task_activation_status":
        "NOT_STARTED",

    "pilot_rebuild_status":
        "NOT_STARTED",

    "question_generation_status":
        "NOT_STARTED",

    "artifact_write_status":
        "NONE",
}


# ============================================================
# 23) Active no-mutation recheck
# ============================================================

cell56_groundability_snapshot_after = hashlib.sha256(
    stable_json_key(
        case_groundability_precheck
    ).encode(
        "utf-8"
    )
).hexdigest()


cell56_design_snapshot_after = hashlib.sha256(
    stable_json_key(
        case_task_design
    ).encode(
        "utf-8"
    )
).hexdigest()


cell56_task_rows_snapshot_after = hashlib.sha256(
    stable_json_key(
        task_draft_plan_rows
    ).encode(
        "utf-8"
    )
).hexdigest()


cell56_task_index_snapshot_after = hashlib.sha256(
    stable_json_key(
        task_draft_plans
    ).encode(
        "utf-8"
    )
).hexdigest()


cell56_ready_snapshot_after = hashlib.sha256(
    stable_json_key(
        ready_case_plans
    ).encode(
        "utf-8"
    )
).hexdigest()


cell56_target_snapshot_after = hashlib.sha256(
    stable_json_key(
        target_evidence_records
    ).encode(
        "utf-8"
    )
).hexdigest()


cell56_pilot_snapshot_after = hashlib.sha256(
    stable_json_key(
        pilot_selection_rows
    ).encode(
        "utf-8"
    )
).hexdigest()


cell56_golden_snapshot_after = hashlib.sha256(
    stable_json_key(
        golden_case_records
    ).encode(
        "utf-8"
    )
).hexdigest()


cell56_case_staging_snapshot_after = hashlib.sha256(
    stable_json_key(
        staged_exercise_v2_golden_case_records
    ).encode(
        "utf-8"
    )
).hexdigest()


# ============================================================
# 24) Validation gates
# ============================================================

legacy_groundability_schema_gate = (
    cell56_groundability_audit[
        "row_count"
    ]
    == 38

    and

    cell56_groundability_audit[
        "schema_uniform"
    ]
)


legacy_task_design_schema_gate = (
    cell56_task_design_audit[
        "row_count"
    ]
    == 38

    and

    cell56_task_design_audit[
        "schema_uniform"
    ]
)


no_unresolved_varying_fields_gate = (
    len(
        cell56_unresolved_groundability_varying_fields
    )
    == 0

    and

    len(
        cell56_unresolved_design_varying_fields
    )
    == 0
)


no_unmapped_stale_constants_gate = (
    len(
        cell56_groundability_stale_constant_failures
    )
    == 0

    and

    len(
        cell56_design_stale_constant_failures
    )
    == 0
)


task_schema_exact_gate = (
    cell56_task_schema_exactly_expected
)


exact_38_groundability_staged_gate = (
    len(
        staged_exercise_v2_groundability_precheck
    )
    == 38

    and

    len(
        cell56_groundability_build_failures
    )
    == 0
)


exact_38_task_design_staged_gate = (
    len(
        staged_exercise_v2_case_task_design
    )
    == 38

    and

    len(
        cell56_task_design_build_failures
    )
    == 0
)


exact_38_task_drafts_staged_gate = (
    len(
        staged_exercise_v2_task_draft_plan_rows
    )
    == 38

    and

    len(
        staged_exercise_v2_task_draft_plans
    )
    == 38

    and

    len(
        cell56_task_draft_build_failures
    )
    == 0
)


exact_case_coverage_gate = (
    cell56_staged_groundability_case_ids
    ==
    cell56_staged_case_ids

    and

    cell56_staged_design_case_ids
    ==
    cell56_staged_case_ids

    and

    cell56_staged_task_case_ids
    ==
    cell56_staged_case_ids
)


task_ids_unique_gate = (
    len(
        cell56_staged_task_ids
    )
    == 38
)


task_ids_recomputable_gate = (
    len(
        cell56_task_id_recomputation_failures
    )
    == 0
)


no_task_id_collision_gate = (
    len(
        cell56_task_id_collisions_with_active
    )
    == 0
)


task_semantics_exact_gate = (
    len(
        cell56_task_semantic_failures
    )
    == 0
)


no_dependency_stale_semantics_gate = (
    len(
        cell56_dependency_stale_semantic_failures
    )
    == 0
)


dependency_case_linkage_gate = (
    len(
        cell56_dependency_case_linkage_failures
    )
    == 0
)


no_pretarget_ready_leakage_gate = (
    cell56_no_pretarget_ready_leakage
)


active_registries_unchanged_gate = (
    cell56_groundability_snapshot_before
    ==
    cell56_groundability_snapshot_after

    and

    cell56_design_snapshot_before
    ==
    cell56_design_snapshot_after

    and

    cell56_task_rows_snapshot_before
    ==
    cell56_task_rows_snapshot_after

    and

    cell56_task_index_snapshot_before
    ==
    cell56_task_index_snapshot_after

    and

    cell56_ready_snapshot_before
    ==
    cell56_ready_snapshot_after

    and

    cell56_target_snapshot_before
    ==
    cell56_target_snapshot_after

    and

    cell56_pilot_snapshot_before
    ==
    cell56_pilot_snapshot_after

    and

    cell56_golden_snapshot_before
    ==
    cell56_golden_snapshot_after

    and

    cell56_case_staging_snapshot_before
    ==
    cell56_case_staging_snapshot_after

    and

    len(
        golden_case_records
    )
    == 724

    and

    len(
        task_draft_plan_rows
    )
    == 724

    and

    len(
        target_evidence_records
    )
    == 572

    and

    len(
        pilot_selection_rows
    )
    == 23
)


target_rebuild_not_started_gate = (
    EXERCISE_V2_TASK_DEPENDENCY_STAGING_V1_DRAFT[
        "target_rebuild_status"
    ]
    ==
    "NOT_STARTED"
)


case_activation_not_started_gate = (
    EXERCISE_V2_TASK_DEPENDENCY_STAGING_V1_DRAFT[
        "case_activation_status"
    ]
    ==
    "NOT_STARTED"
)


task_activation_not_started_gate = (
    EXERCISE_V2_TASK_DEPENDENCY_STAGING_V1_DRAFT[
        "task_activation_status"
    ]
    ==
    "NOT_STARTED"
)


pilot_rebuild_not_started_gate = (
    EXERCISE_V2_TASK_DEPENDENCY_STAGING_V1_DRAFT[
        "pilot_rebuild_status"
    ]
    ==
    "NOT_STARTED"
)


question_generation_not_started_gate = (
    EXERCISE_V2_TASK_DEPENDENCY_STAGING_V1_DRAFT[
        "question_generation_status"
    ]
    ==
    "NOT_STARTED"
)


no_artifact_write_gate = (
    EXERCISE_V2_TASK_DEPENDENCY_STAGING_V1_DRAFT[
        "artifact_write_status"
    ]
    ==
    "NONE"
)


cell_56_gates = {
    "legacy_groundability_schema_uniform_for_exact_38_cases":
        legacy_groundability_schema_gate,

    "legacy_task_design_schema_uniform_for_exact_38_cases":
        legacy_task_design_schema_gate,

    "no_varying_dependency_field_lacks_explicit_v2_derivation":
        no_unresolved_varying_fields_gate,

    "no_unmapped_stale_legacy_semantic_constant_would_be_preserved":
        no_unmapped_stale_constants_gate,

    "task_draft_schema_exactly_matches_actual_notebook_contract":
        task_schema_exact_gate,

    "exact_38_v2_groundability_records_staged":
        exact_38_groundability_staged_gate,

    "exact_38_v2_case_task_design_records_staged":
        exact_38_task_design_staged_gate,

    "exact_38_v2_task_draft_records_staged":
        exact_38_task_drafts_staged_gate,

    "all_three_staged_dependency_families_exactly_cover_the_38_v2_cases":
        exact_case_coverage_gate,

    "all_38_v2_task_draft_ids_unique":
        task_ids_unique_gate,

    "all_38_v2_task_draft_ids_independently_recomputable":
        task_ids_recomputable_gate,

    "no_v2_task_draft_id_collides_with_current_724_tasks":
        no_task_id_collision_gate,

    "all_38_staged_task_semantics_match_repaired_v2_pretarget_contract":
        task_semantics_exact_gate,

    "no_staged_groundability_or_task_design_record_carries_stale_legacy_semantics":
        no_dependency_stale_semantics_gate,

    "all_staged_dependencies_link_only_to_staged_v2_case_ids":
        dependency_case_linkage_gate,

    "no_v2_task_leaks_into_ready_case_plans_before_target_validation":
        no_pretarget_ready_leakage_gate,

    "all_active_case_task_target_pilot_registries_content_unchanged":
        active_registries_unchanged_gate,

    "target_rebuild_not_started":
        target_rebuild_not_started_gate,

    "case_activation_not_started":
        case_activation_not_started_gate,

    "task_activation_not_started":
        task_activation_not_started_gate,

    "pilot_rebuild_not_started":
        pilot_rebuild_not_started_gate,

    "question_generation_not_started":
        question_generation_not_started_gate,

    "no_artifact_write":
        no_artifact_write_gate,
}


cell_56_pass = all(
    cell_56_gates.values()
)


# ============================================================
# 25) Output
# ============================================================

print(
    "\n=== EXERCISE V2 TASK DEPENDENCY STAGING ==="
)


print(
    "Dependency schema safe to stage:",
    cell56_dependency_schema_safe_to_stage
)


print(
    "\nUnresolved Groundability varying fields:",
    cell56_unresolved_groundability_varying_fields
)


print(
    "Unresolved Task Design varying fields:",
    cell56_unresolved_design_varying_fields
)


print(
    "Groundability stale constants:",
    cell56_groundability_stale_constant_failures
)


print(
    "Task Design stale constants:",
    cell56_design_stale_constant_failures
)


print(
    "\n=== STAGED COUNTS ==="
)


print(
    "Groundability:",
    len(
        staged_exercise_v2_groundability_precheck
    )
)


print(
    "Task Design:",
    len(
        staged_exercise_v2_case_task_design
    )
)


print(
    "Task Drafts:",
    len(
        staged_exercise_v2_task_draft_plan_rows
    )
)


print(
    "Ready Case Plans:",
    len(
        staged_exercise_v2_ready_case_plans
    )
)


print(
    "\n=== BUILD FAILURES ==="
)


print(
    "Groundability failures:",
    len(
        cell56_groundability_build_failures
    )
)


print(
    "Task Design failures:",
    len(
        cell56_task_design_build_failures
    )
)


print(
    "Task Draft failures:",
    len(
        cell56_task_draft_build_failures
    )
)


if (
    cell56_groundability_build_failures
):

    print(
        "\nFirst Groundability failures:"
    )

    for row in (
        cell56_groundability_build_failures[
            :10
        ]
    ):

        print(
            "-",
            row
        )


if (
    cell56_task_design_build_failures
):

    print(
        "\nFirst Task Design failures:"
    )

    for row in (
        cell56_task_design_build_failures[
            :10
        ]
    ):

        print(
            "-",
            row
        )


if (
    cell56_task_draft_build_failures
):

    print(
        "\nFirst Task Draft failures:"
    )

    for row in (
        cell56_task_draft_build_failures[
            :10
        ]
    ):

        print(
            "-",
            row
        )


print(
    "\n=== TASK-ID VALIDATION ==="
)


print(
    "Unique staged Task IDs:",
    len(
        cell56_staged_task_ids
    )
)


print(
    "Task-ID recomputation failures:",
    len(
        cell56_task_id_recomputation_failures
    )
)


print(
    "Task-ID collisions with active tasks:",
    len(
        cell56_task_id_collisions_with_active
    )
)


print(
    "\n=== SEMANTIC VALIDATION ==="
)


print(
    "Task semantic failures:",
    len(
        cell56_task_semantic_failures
    )
)


print(
    "Dependency stale-semantic failures:",
    len(
        cell56_dependency_stale_semantic_failures
    )
)


print(
    "Dependency linkage failures:",
    len(
        cell56_dependency_case_linkage_failures
    )
)


print(
    "Pre-target Ready leakage:",
    not cell56_no_pretarget_ready_leakage
)


print(
    "\nStaging Snapshot SHA256:",
    cell56_staging_snapshot_sha256
)


print(
    "\n=== FIRST 3 STAGED TASK DRAFTS ==="
)


for task_row in (
    staged_exercise_v2_task_draft_plan_rows[
        :3
    ]
):

    print(
        "\nTask:",
        task_row[
            "task_draft_id"
        ]
    )

    print(
        "Case:",
        task_row[
            "golden_case_id"
        ]
    )

    print(
        "Family:",
        task_row[
            "task_family"
        ]
    )

    print(
        "Scope:",
        task_row[
            "evidence_scope"
        ]
    )

    print(
        "Method:",
        task_row[
            "evidence_target_method"
        ]
    )

    print(
        "Construction:",
        task_row[
            "construction_eligibility"
        ]
    )

    print(
        "Target status:",
        task_row[
            "target_evidence_status"
        ]
    )


print(
    "\n=== ACTIVE COUNTS ==="
)


print(
    "Golden:",
    len(
        golden_case_records
    )
)


print(
    "Tasks:",
    len(
        task_draft_plan_rows
    )
)


print(
    "Targets:",
    len(
        target_evidence_records
    )
)


print(
    "Pilots:",
    len(
        pilot_selection_rows
    )
)


print(
    "\n=== CODE CELL 56 GATES ==="
)


for name, passed in (
    cell_56_gates.items()
):

    print(
        name,
        ":",
        "PASS"
        if passed
        else "FAIL"
    )


print(
    "\nCODE CELL 56 GATE:",
    "PASS"
    if cell_56_pass
    else "FAIL"
)


assert cell_56_pass, (
    "Exercise V2 Task dependency staging failed. "
    "Do not rebuild Targets or activate V2 cases. "
    "Repair this SAME Code Cell 56 until Groundability, "
    "Case Task Design and Task Draft staging all exactly cover "
    "the 38 V2 cases without stale semantics, unknown varying "
    "fields, ID collisions or mutation of active registries."
)
```

```python
# Code Cell 57 — REPAIRED AGAIN

# ============================================================
# Legacy Exercise Target Evidence Contract + Stable-ID Discovery
#
# REPAIR:
#
# Previous repaired run failed inside Target-ID value-shape
# discovery because uniqueness was computed with:
#
#     set(values)
#
# Some Target fields contain lists/dicts, so they are not
# hashable.
#
# This version measures uniqueness using canonical serialized
# values:
#
#     stable_json_key(value)
#
# Therefore every Target field can be inspected safely without
# assuming it is a scalar.
#
# READ-ONLY.
#
# NO V2 Target creation.
# NO Case activation.
# NO Task activation.
# NO Pilot rebuild.
# NO question generation.
# NO artifact writes.
# ============================================================


# ============================================================
# 1) Freeze namespace and active/staging snapshots
# ============================================================

cell57_globals_snapshot = dict(
    globals()
)

cell57_globals_snapshot_items = list(
    cell57_globals_snapshot.items()
)


cell57_target_snapshot_before = hashlib.sha256(
    stable_json_key(
        target_evidence_records
    ).encode(
        "utf-8"
    )
).hexdigest()


cell57_target_by_case_snapshot_before = hashlib.sha256(
    stable_json_key(
        target_evidence_by_case_id
    ).encode(
        "utf-8"
    )
).hexdigest()


cell57_target_by_task_snapshot_before = hashlib.sha256(
    stable_json_key(
        target_evidence_by_task_draft_id
    ).encode(
        "utf-8"
    )
).hexdigest()


cell57_task_snapshot_before = hashlib.sha256(
    stable_json_key(
        task_draft_plan_rows
    ).encode(
        "utf-8"
    )
).hexdigest()


cell57_golden_snapshot_before = hashlib.sha256(
    stable_json_key(
        golden_case_records
    ).encode(
        "utf-8"
    )
).hexdigest()


cell57_pilot_snapshot_before = hashlib.sha256(
    stable_json_key(
        pilot_selection_rows
    ).encode(
        "utf-8"
    )
).hexdigest()


cell57_v2_case_staging_snapshot_before = hashlib.sha256(
    stable_json_key(
        staged_exercise_v2_golden_case_records
    ).encode(
        "utf-8"
    )
).hexdigest()


cell57_v2_task_staging_snapshot_before = hashlib.sha256(
    stable_json_key(
        staged_exercise_v2_task_draft_plan_rows
    ).encode(
        "utf-8"
    )
).hexdigest()


# ============================================================
# 2) Exact 38 Legacy Exercise Target records
# ============================================================

cell57_legacy_target_rows = sorted(
    [
        row

        for row
        in target_evidence_records

        if (
            row[
                "golden_case_id"
            ]
            in
            cell46_legacy_exercise_case_ids
        )
    ],
    key=lambda row:
        row[
            "golden_case_id"
        ],
)


# ============================================================
# 3) Uniform Target schema
# ============================================================

cell57_target_key_sets = [
    set(
        row.keys()
    )

    for row
    in cell57_legacy_target_rows
]


cell57_target_key_union = (
    set().union(
        *cell57_target_key_sets
    )
    if cell57_target_key_sets
    else set()
)


cell57_target_key_intersection = (
    set(
        cell57_target_key_sets[
            0
        ]
    )
    if cell57_target_key_sets
    else set()
)


for key_set in (
    cell57_target_key_sets[
        1:
    ]
):

    cell57_target_key_intersection &= (
        key_set
    )


cell57_target_schema_uniform = (
    cell57_target_key_union
    ==
    cell57_target_key_intersection
)


# ============================================================
# 4) Field distinct-count / constant-value audit
# ============================================================

cell57_target_field_distinct_counts = {}

cell57_target_constant_fields = {}


for field_name in sorted(
    cell57_target_key_union
):

    values = [
        row.get(
            field_name
        )

        for row
        in cell57_legacy_target_rows
    ]


    serialized_values = [
        stable_json_key(
            value
        )

        for value
        in values
    ]


    distinct_values = set(
        serialized_values
    )


    cell57_target_field_distinct_counts[
        field_name
    ] = len(
        distinct_values
    )


    if (
        len(
            distinct_values
        )
        == 1
    ):

        cell57_target_constant_fields[
            field_name
        ] = values[
            0
        ]


cell57_target_varying_fields = sorted(
    [
        field_name

        for field_name, count
        in cell57_target_field_distinct_counts.items()

        if (
            count
            > 1
        )
    ]
)


# ============================================================
# 5) Exact Case -> Task -> Target linkage
# ============================================================

cell57_legacy_task_by_case_id = {
    row[
        "golden_case_id"
    ]:
        row

    for row
    in cell54_legacy_task_rows
}


cell57_legacy_target_by_case_id = {
    row[
        "golden_case_id"
    ]:
        row

    for row
    in cell57_legacy_target_rows
}


cell57_case_task_target_linkage_failures = []


for legacy_case_id in sorted(
    cell46_legacy_exercise_case_ids
):

    task_row = (
        cell57_legacy_task_by_case_id.get(
            legacy_case_id
        )
    )


    target_row = (
        cell57_legacy_target_by_case_id.get(
            legacy_case_id
        )
    )


    if (
        task_row
        is None
    ):

        cell57_case_task_target_linkage_failures.append(
            {
                "golden_case_id":
                    legacy_case_id,

                "reason":
                    "TASK_ROW_MISSING",
            }
        )

        continue


    if (
        target_row
        is None
    ):

        cell57_case_task_target_linkage_failures.append(
            {
                "golden_case_id":
                    legacy_case_id,

                "reason":
                    "TARGET_ROW_MISSING",
            }
        )

        continue


    if (
        target_row[
            "task_draft_id"
        ]
        !=
        task_row[
            "task_draft_id"
        ]
    ):

        cell57_case_task_target_linkage_failures.append(
            {
                "golden_case_id":
                    legacy_case_id,

                "reason":
                    "TARGET_TASK_ID_MISMATCH",

                "task_draft_id":
                    task_row[
                        "task_draft_id"
                    ],

                "target_task_draft_id":
                    target_row[
                        "task_draft_id"
                    ],
            }
        )


# ============================================================
# 6) Validate Target indexes
# ============================================================

cell57_target_case_index_failures = []

cell57_target_task_index_failures = []


for target_row in (
    cell57_legacy_target_rows
):

    case_id = (
        target_row[
            "golden_case_id"
        ]
    )


    task_id = (
        target_row[
            "task_draft_id"
        ]
    )


    if (
        case_id
        not in
        target_evidence_by_case_id
    ):

        cell57_target_case_index_failures.append(
            {
                "golden_case_id":
                    case_id,

                "reason":
                    "MISSING_FROM_CASE_INDEX",
            }
        )


    elif (
        stable_json_key(
            target_evidence_by_case_id[
                case_id
            ]
        )
        !=
        stable_json_key(
            target_row
        )
    ):

        cell57_target_case_index_failures.append(
            {
                "golden_case_id":
                    case_id,

                "reason":
                    "CASE_INDEX_CONTENT_MISMATCH",
            }
        )


    if (
        task_id
        not in
        target_evidence_by_task_draft_id
    ):

        cell57_target_task_index_failures.append(
            {
                "task_draft_id":
                    task_id,

                "reason":
                    "MISSING_FROM_TASK_INDEX",
            }
        )


    elif (
        stable_json_key(
            target_evidence_by_task_draft_id[
                task_id
            ]
        )
        !=
        stable_json_key(
            target_row
        )
    ):

        cell57_target_task_index_failures.append(
            {
                "task_draft_id":
                    task_id,

                "reason":
                    "TASK_INDEX_CONTENT_MISMATCH",
            }
        )


# ============================================================
# 7) Resolve Target-ID field from ACTUAL VALUE SHAPE
#
# IMPORTANT REPAIR:
#   uniqueness is calculated from canonical serialization,
#   therefore list/dict-valued fields are safe to inspect.
# ============================================================

cell57_target_id_value_shape_candidates = []


cell57_target_id_value_shape_profiles = {}


for field_name in sorted(
    cell57_target_key_union
):

    values = [
        row[
            field_name
        ]

        for row
        in cell57_legacy_target_rows
    ]


    serialized_values = [
        stable_json_key(
            value
        )

        for value
        in values
    ]


    all_strings = all(
        isinstance(
            value,
            str,
        )

        for value
        in values
    )


    all_target_id_shape = (
        all_strings

        and

        all(
            re.fullmatch(
                r"target-evidence:[0-9a-f]{32}",
                value,
            )
            is not None

            for value
            in values
        )
    )


    all_unique = (
        len(
            set(
                serialized_values
            )
        )
        ==
        38
    )


    cell57_target_id_value_shape_profiles[
        field_name
    ] = {
        "all_strings":
            all_strings,

        "all_target_id_shape":
            all_target_id_shape,

        "distinct_count":
            len(
                set(
                    serialized_values
                )
            ),

        "all_unique":
            all_unique,
    }


    if (
        all_target_id_shape
        and
        all_unique
    ):

        cell57_target_id_value_shape_candidates.append(
            field_name
        )


cell57_target_id_field_name = (
    cell57_target_id_value_shape_candidates[
        0
    ]
    if (
        len(
            cell57_target_id_value_shape_candidates
        )
        == 1
    )
    else None
)


# ============================================================
# 8) Source-level Target formula evidence
# ============================================================

CELL57_SOURCE_TERMS = [
    "def build_target_evidence_id",
    "target-evidence:",
    "target_part_identity",
    "TARGET_EVIDENCE_SCHEMA_VERSION",
]


cell57_candidate_source_cells = []


for code_cell in (
    cell49_code_cells
):

    source = (
        code_cell[
            "source"
        ]
    )


    matched_terms = [
        term

        for term
        in CELL57_SOURCE_TERMS

        if (
            term
            in
            source
        )
    ]


    if not (
        matched_terms
    ):

        continue


    if not (
        "def build_target_evidence_id"
        in
        source

        or

        "target-evidence:"
        in
        source
    ):

        continue


    cell57_candidate_source_cells.append(
        {
            "cell_index":
                code_cell[
                    "cell_index"
                ],

            "matched_terms":
                matched_terms,

            "source":
                source,
        }
    )


cell57_formula_evidence_snippets = []


for source_row in (
    cell57_candidate_source_cells
):

    source_lines = (
        source_row[
            "source"
        ].splitlines()
    )


    for matched_index, line in enumerate(
        source_lines
    ):

        if not (
            "def build_target_evidence_id"
            in
            line

            or

            "target-evidence:"
            in
            line

            or

            "target_part_identity"
            in
            line
        ):

            continue


        start_index = max(
            0,
            matched_index
            - 10,
        )


        end_index = min(
            len(
                source_lines
            ),
            matched_index
            + 28,
        )


        cell57_formula_evidence_snippets.append(
            {
                "cell_index":
                    source_row[
                        "cell_index"
                    ],

                "matched_line_index":
                    matched_index,

                "snippet":
                    "\n".join(
                        source_lines[
                            start_index:
                            end_index
                        ]
                    ),
            }
        )


# ============================================================
# 9) Resolve actual Target-ID builder explicitly
# ============================================================

cell57_build_target_evidence_id_callable = (
    cell57_globals_snapshot.get(
        "build_target_evidence_id"
    )
)


cell57_target_part_identity_callable = (
    cell57_globals_snapshot.get(
        "target_part_identity"
    )
)


cell57_actual_builder_present = callable(
    cell57_build_target_evidence_id_callable
)


cell57_target_part_identity_present = callable(
    cell57_target_part_identity_callable
)


# ============================================================
# 10) Validate actual Builder signature
# ============================================================

cell57_builder_signature = None


if (
    cell57_actual_builder_present
):

    cell57_builder_code = (
        cell57_build_target_evidence_id_callable.__code__
    )


    cell57_builder_signature = list(
        cell57_builder_code.co_varnames[
            :cell57_builder_code.co_argcount
        ]
    )


cell57_expected_builder_signature = [
    "task_draft_id",
    "method",
    "parts",
]


cell57_builder_signature_exact = (
    cell57_builder_signature
    ==
    cell57_expected_builder_signature
)


# ============================================================
# 11) Explicit persisted-record -> builder-argument mapping
# ============================================================

CELL57_TARGET_BUILDER_ARGUMENT_MAPPING = {
    "task_draft_id":
        "task_draft_id",

    "method":
        "evidence_target_method",

    "parts":
        "target_parts",
}


cell57_builder_argument_mapping_fields_present = all(
    persisted_field
    in
    cell57_target_key_union

    for persisted_field
    in CELL57_TARGET_BUILDER_ARGUMENT_MAPPING.values()
)


# ============================================================
# 12) Reproduce all 38 IDs using actual builder
# ============================================================

cell57_builder_reproduction_failures = []


if (
    cell57_actual_builder_present
    and
    cell57_builder_signature_exact
    and
    cell57_builder_argument_mapping_fields_present
    and
    cell57_target_id_field_name
    is not None
):

    for target_row in (
        cell57_legacy_target_rows
    ):

        try:

            reproduced_target_id = (
                cell57_build_target_evidence_id_callable(
                    task_draft_id=(
                        target_row[
                            "task_draft_id"
                        ]
                    ),
                    method=(
                        target_row[
                            "evidence_target_method"
                        ]
                    ),
                    parts=(
                        target_row[
                            "target_parts"
                        ]
                    ),
                )
            )


        except Exception as exc:

            cell57_builder_reproduction_failures.append(
                {
                    "golden_case_id":
                        target_row[
                            "golden_case_id"
                        ],

                    "reason":
                        "BUILDER_EXCEPTION",

                    "error_type":
                        type(
                            exc
                        ).__name__,

                    "error_message":
                        str(
                            exc
                        ),
                }
            )

            continue


        if (
            reproduced_target_id
            !=
            target_row[
                cell57_target_id_field_name
            ]
        ):

            cell57_builder_reproduction_failures.append(
                {
                    "golden_case_id":
                        target_row[
                            "golden_case_id"
                        ],

                    "reason":
                        "ID_MISMATCH",

                    "stored":
                        target_row[
                            cell57_target_id_field_name
                        ],

                    "reproduced":
                        reproduced_target_id,
                }
            )


cell57_builder_exact_match_count = (
    38
    -
    len(
        cell57_builder_reproduction_failures
    )
)


# ============================================================
# 13) Independently reproduce all Target IDs
#
# Does NOT call build_target_evidence_id.
# ============================================================

cell57_independent_target_id_failures = []


if (
    cell57_target_part_identity_present

    and

    cell57_target_id_field_name
    is not None
):

    for target_row in (
        cell57_legacy_target_rows
    ):

        independent_payload = {
            "schema_version":
                TARGET_EVIDENCE_SCHEMA_VERSION,

            "task_draft_id":
                target_row[
                    "task_draft_id"
                ],

            "method":
                target_row[
                    "evidence_target_method"
                ],

            "parts":
                [
                    cell57_target_part_identity_callable(
                        part
                    )

                    for part
                    in target_row[
                        "target_parts"
                    ]
                ],
        }


        independent_digest = hashlib.sha256(
            stable_json_key(
                independent_payload
            ).encode(
                "utf-8"
            )
        ).hexdigest()


        independent_target_id = (
            "target-evidence:"
            +
            independent_digest[
                :32
            ]
        )


        if (
            independent_target_id
            !=
            target_row[
                cell57_target_id_field_name
            ]
        ):

            cell57_independent_target_id_failures.append(
                {
                    "golden_case_id":
                        target_row[
                            "golden_case_id"
                        ],

                    "stored":
                        target_row[
                            cell57_target_id_field_name
                        ],

                    "recomputed":
                        independent_target_id,
                }
            )


# ============================================================
# 14) Formula resolution
# ============================================================

cell57_target_id_formula_status = (
    "RESOLVED_BY_EXISTING_BUILDER_AND_INDEPENDENT_RECOMPUTATION_38_OF_38"
    if (
        cell57_target_id_field_name
        ==
        "target_evidence_id"

        and

        cell57_actual_builder_present

        and

        cell57_builder_signature_exact

        and

        len(
            cell57_builder_reproduction_failures
        )
        == 0

        and

        len(
            cell57_independent_target_id_failures
        )
        == 0
    )
    else
    "NOT_YET_RESOLVED"
)


cell57_resolved_target_id_builder_name = (
    "build_target_evidence_id"
    if (
        cell57_target_id_formula_status
        ==
        "RESOLVED_BY_EXISTING_BUILDER_AND_INDEPENDENT_RECOMPUTATION_38_OF_38"
    )
    else None
)


# ============================================================
# 15) Target Part schema/profile audit
# ============================================================

cell57_target_parts_length_distribution = Counter(
    len(
        row[
            "target_parts"
        ]
    )

    for row
    in cell57_legacy_target_rows
)


cell57_target_text_spans_length_distribution = Counter(
    len(
        row[
            "target_text_spans"
        ]
    )

    for row
    in cell57_legacy_target_rows
)


cell57_target_part_key_sets = [
    set(
        part.keys()
    )

    for row
    in cell57_legacy_target_rows

    for part
    in row[
        "target_parts"
    ]
]


cell57_target_part_key_union = (
    set().union(
        *cell57_target_part_key_sets
    )
    if cell57_target_part_key_sets
    else set()
)


cell57_target_part_key_intersection = (
    set(
        cell57_target_part_key_sets[
            0
        ]
    )
    if cell57_target_part_key_sets
    else set()
)


for key_set in (
    cell57_target_part_key_sets[
        1:
    ]
):

    cell57_target_part_key_intersection &= (
        key_set
    )


cell57_target_part_schema_uniform = (
    cell57_target_part_key_union
    ==
    cell57_target_part_key_intersection
)


# ============================================================
# 16) Verify persisted target_parts drive non-empty identity
# ============================================================

cell57_part_identity_recompute_failures = []


if (
    cell57_target_part_identity_present
):

    for target_row in (
        cell57_legacy_target_rows
    ):

        try:

            normalized_parts = [
                cell57_target_part_identity_callable(
                    part
                )

                for part
                in target_row[
                    "target_parts"
                ]
            ]


        except Exception as exc:

            cell57_part_identity_recompute_failures.append(
                {
                    "golden_case_id":
                        target_row[
                            "golden_case_id"
                        ],

                    "reason":
                        "TARGET_PART_IDENTITY_EXCEPTION",

                    "error_type":
                        type(
                            exc
                        ).__name__,

                    "error_message":
                        str(
                            exc
                        ),
                }
            )

            continue


        if not (
            normalized_parts
        ):

            cell57_part_identity_recompute_failures.append(
                {
                    "golden_case_id":
                        target_row[
                            "golden_case_id"
                        ],

                    "reason":
                        "EMPTY_NORMALIZED_PART_IDENTITY",
                }
            )


# ============================================================
# 17) Exact Document / Span linkage audit
# ============================================================

cell57_target_document_linkage_failures = []

cell57_target_span_linkage_failures = []


for target_row in (
    cell57_legacy_target_rows
):

    case_id = (
        target_row[
            "golden_case_id"
        ]
    )


    case_record = (
        golden_case_records_by_id[
            case_id
        ]
    )


    case_evidence_ids = set(
        case_record[
            "evidence_document_ids"
        ]
    )


    target_document_ids = list(
        target_row[
            "target_document_ids"
        ]
    )


    if not (
        set(
            target_document_ids
        ).issubset(
            case_evidence_ids
        )
    ):

        cell57_target_document_linkage_failures.append(
            {
                "golden_case_id":
                    case_id,

                "target_document_ids":
                    target_document_ids,

                "case_evidence_document_ids":
                    sorted(
                        case_evidence_ids
                    ),
            }
        )


    for span in (
        target_row[
            "target_text_spans"
        ]
    ):

        document_id = (
            span[
                "document_id"
            ]
        )


        if (
            document_id
            not in
            target_document_ids
        ):

            cell57_target_span_linkage_failures.append(
                {
                    "golden_case_id":
                        case_id,

                    "document_id":
                        document_id,

                    "reason":
                        "SPAN_DOCUMENT_NOT_IN_TARGET_DOCUMENT_IDS",
                }
            )

            continue


        if (
            document_id
            not in
            corpus_target_index
        ):

            cell57_target_span_linkage_failures.append(
                {
                    "golden_case_id":
                        case_id,

                    "document_id":
                        document_id,

                    "reason":
                        "SPAN_DOCUMENT_NOT_IN_CORPUS_TARGET_INDEX",
                }
            )

            continue


        corpus_record = (
            corpus_target_index[
                document_id
            ]
        )


        text = (
            corpus_record[
                "text"
            ]
        )


        char_start = (
            span[
                "char_start"
            ]
        )


        char_end = (
            span[
                "char_end"
            ]
        )


        valid_bounds = (
            isinstance(
                char_start,
                int,
            )

            and

            isinstance(
                char_end,
                int,
            )

            and

            0
            <=
            char_start
            <
            char_end
            <=
            len(
                text
            )
        )


        if not (
            valid_bounds
        ):

            cell57_target_span_linkage_failures.append(
                {
                    "golden_case_id":
                        case_id,

                    "document_id":
                        document_id,

                    "reason":
                        "INVALID_CHAR_BOUNDS",
                }
            )

            continue


        target_text = (
            text[
                char_start:
                char_end
            ]
        )


        recomputed_text_sha = hashlib.sha256(
            target_text.encode(
                "utf-8"
            )
        ).hexdigest()


        if (
            recomputed_text_sha
            !=
            span[
                "target_text_sha256"
            ]
        ):

            cell57_target_span_linkage_failures.append(
                {
                    "golden_case_id":
                        case_id,

                    "document_id":
                        document_id,

                    "reason":
                        "TARGET_TEXT_SHA_MISMATCH",
                }
            )


# ============================================================
# 18) Legacy Target evidence shape summary
# ============================================================

cell57_target_document_count_distribution = Counter(
    len(
        row[
            "target_document_ids"
        ]
    )

    for row
    in cell57_legacy_target_rows
)


cell57_target_total_tokens_distribution = Counter(
    row[
        "target_total_tokens"
    ]

    for row
    in cell57_legacy_target_rows
)


# ============================================================
# 19) Explicit V2 Target field-handling policy
# ============================================================

cell57_v2_target_field_policy = {
    "target_evidence_id":
        "RECOMPUTE_FROM_V2_TASK_METHOD_AND_PARTS",

    "target_schema_version":
        "REVIEW_SCHEMA_COMPATIBILITY_BEFORE_V2_TARGET_BUILD",

    "task_draft_id":
        "USE_STAGED_V2_TASK_DRAFT_ID",

    "golden_case_id":
        "USE_STAGED_V2_GOLDEN_CASE_ID",

    "source_type":
        "PRESERVE_BOOK",

    "stratum":
        "PRESERVE_EXERCISE_OR_SELF_ASSESSMENT",

    "evidence_target_method":
        "USE_V2_EXPLICIT_BOOK_EXERCISE_UNIT_METHOD",

    "target_document_ids":
        "REBUILD_FROM_COMPLETE_SELECTED_EXERCISE_UNIT_EVIDENCE",

    "target_parts":
        "REBUILD_FOR_V2_EXERCISE_UNIT",

    "target_text_spans":
        "REBUILD_FROM_V2_TARGET_PARTS",

    "target_locator_refs":
        "REBUILD_FROM_V2_TARGET_DOCUMENTS",

    "target_total_tokens":
        "RECOMPUTE_FROM_V2_TARGET_PARTS",

    "target_selection_basis":
        "DEFINE_V2_EXERCISE_UNIT_SELECTION_BASIS",

    "target_attribution_state":
        "REBUILD_FROM_V2_TARGET_DOCUMENT_EVIDENCE",

    "target_authority_state":
        "REBUILD_WITHOUT_AUTHORITY_INFERENCE",

    "question_generation_status":
        "PRESERVE_NOT_STARTED",

    "expected_answer_generation_status":
        "PRESERVE_NOT_STARTED",

    "citation_target_status":
        "REBUILD_LOCATOR_CAPTURE_STATUS_AFTER_V2_TARGET_BUILD",
}


cell57_unclassified_target_fields = sorted(
    cell57_target_key_union
    -
    set(
        cell57_v2_target_field_policy.keys()
    )
)


cell57_unknown_policy_fields = sorted(
    set(
        cell57_v2_target_field_policy.keys()
    )
    -
    cell57_target_key_union
)


# ============================================================
# 20) Important V2 semantic decision remains OPEN
#
# Legacy:
#   38/38 targets = 1 target part + 1 text span.
#
# V2:
#   Exercise Unit evidence = 2 .. 153 documents.
#
# We intentionally do NOT yet decide whether V2 target_parts
# should be:
#   - every document,
#   - an aggregate,
#   - a deterministic subset,
#   - or another structural representation.
# ============================================================

cell57_v2_target_part_strategy_status = (
    "NOT_YET_DECIDED_REQUIRES_EXERCISE_UNIT_TARGET_DESIGN"
)


# ============================================================
# 21) Contract discovery registry
# ============================================================

EXERCISE_V2_TARGET_CONTRACT_DISCOVERY_V1_DRAFT = {
    "version":
        "exercise_v2_target_contract_discovery_v1_draft",

    "status":
        (
            "LEGACY_TARGET_CONTRACT_AND_ID_FORMULA_RESOLVED"
            if (
                cell57_target_id_formula_status
                ==
                "RESOLVED_BY_EXISTING_BUILDER_AND_INDEPENDENT_RECOMPUTATION_38_OF_38"
            )
            else
            "TARGET_ID_FORMULA_NOT_YET_RESOLVED"
        ),

    "legacy_exercise_target_count":
        len(
            cell57_legacy_target_rows
        ),

    "legacy_target_schema_uniform":
        cell57_target_schema_uniform,

    "target_id_field":
        cell57_target_id_field_name,

    "target_id_value_shape":
        "target-evidence:<32 lowercase hex>",

    "target_id_formula_status":
        cell57_target_id_formula_status,

    "resolved_target_id_builder_name":
        cell57_resolved_target_id_builder_name,

    "builder_argument_mapping":
        dict(
            CELL57_TARGET_BUILDER_ARGUMENT_MAPPING
        ),

    "legacy_target_parts_length_distribution":
        dict(
            cell57_target_parts_length_distribution
        ),

    "legacy_target_text_spans_length_distribution":
        dict(
            cell57_target_text_spans_length_distribution
        ),

    "legacy_target_document_count_distribution":
        dict(
            cell57_target_document_count_distribution
        ),

    "v2_target_part_strategy_status":
        cell57_v2_target_part_strategy_status,

    "v2_target_staging_status":
        "NOT_STARTED",

    "case_activation_status":
        "NOT_STARTED",

    "task_activation_status":
        "NOT_STARTED",

    "pilot_rebuild_status":
        "NOT_STARTED",

    "question_generation_status":
        "NOT_STARTED",

    "artifact_write_status":
        "NONE",
}


# ============================================================
# 22) No-mutation recheck
# ============================================================

cell57_target_snapshot_after = hashlib.sha256(
    stable_json_key(
        target_evidence_records
    ).encode(
        "utf-8"
    )
).hexdigest()


cell57_target_by_case_snapshot_after = hashlib.sha256(
    stable_json_key(
        target_evidence_by_case_id
    ).encode(
        "utf-8"
    )
).hexdigest()


cell57_target_by_task_snapshot_after = hashlib.sha256(
    stable_json_key(
        target_evidence_by_task_draft_id
    ).encode(
        "utf-8"
    )
).hexdigest()


cell57_task_snapshot_after = hashlib.sha256(
    stable_json_key(
        task_draft_plan_rows
    ).encode(
        "utf-8"
    )
).hexdigest()


cell57_golden_snapshot_after = hashlib.sha256(
    stable_json_key(
        golden_case_records
    ).encode(
        "utf-8"
    )
).hexdigest()


cell57_pilot_snapshot_after = hashlib.sha256(
    stable_json_key(
        pilot_selection_rows
    ).encode(
        "utf-8"
    )
).hexdigest()


cell57_v2_case_staging_snapshot_after = hashlib.sha256(
    stable_json_key(
        staged_exercise_v2_golden_case_records
    ).encode(
        "utf-8"
    )
).hexdigest()


cell57_v2_task_staging_snapshot_after = hashlib.sha256(
    stable_json_key(
        staged_exercise_v2_task_draft_plan_rows
    ).encode(
        "utf-8"
    )
).hexdigest()


# ============================================================
# 23) Validation gates
# ============================================================

exact_38_targets_gate = (
    len(
        cell57_legacy_target_rows
    )
    == 38
)


uniform_target_schema_gate = (
    cell57_target_schema_uniform
)


exact_case_task_target_linkage_gate = (
    len(
        cell57_case_task_target_linkage_failures
    )
    == 0
)


target_case_index_gate = (
    len(
        cell57_target_case_index_failures
    )
    == 0
)


target_task_index_gate = (
    len(
        cell57_target_task_index_failures
    )
    == 0
)


exactly_one_value_shape_target_id_field_gate = (
    len(
        cell57_target_id_value_shape_candidates
    )
    == 1

    and

    cell57_target_id_field_name
    ==
    "target_evidence_id"
)


source_formula_evidence_found_gate = (
    len(
        cell57_candidate_source_cells
    )
    > 0

    and

    len(
        cell57_formula_evidence_snippets
    )
    > 0
)


actual_target_builder_present_gate = (
    cell57_actual_builder_present
    and
    cell57_target_part_identity_present
)


actual_target_builder_signature_gate = (
    cell57_builder_signature_exact
)


builder_argument_mapping_grounded_gate = (
    cell57_builder_argument_mapping_fields_present

    and

    CELL57_TARGET_BUILDER_ARGUMENT_MAPPING
    ==
    {
        "task_draft_id":
            "task_draft_id",

        "method":
            "evidence_target_method",

        "parts":
            "target_parts",
    }
)


actual_builder_reproduces_38_gate = (
    cell57_builder_exact_match_count
    == 38

    and

    len(
        cell57_builder_reproduction_failures
    )
    == 0
)


independent_target_id_reproduction_gate = (
    len(
        cell57_independent_target_id_failures
    )
    == 0
)


target_id_formula_resolved_gate = (
    cell57_target_id_formula_status
    ==
    "RESOLVED_BY_EXISTING_BUILDER_AND_INDEPENDENT_RECOMPUTATION_38_OF_38"
)


legacy_parts_are_exactly_single_part_gate = (
    cell57_target_parts_length_distribution
    ==
    Counter(
        {
            1:
                38
        }
    )
)


legacy_spans_are_exactly_single_span_gate = (
    cell57_target_text_spans_length_distribution
    ==
    Counter(
        {
            1:
                38
        }
    )
)


target_part_schema_uniform_gate = (
    cell57_target_part_schema_uniform
)


target_part_identity_nonempty_gate = (
    len(
        cell57_part_identity_recompute_failures
    )
    == 0
)


target_document_linkage_gate = (
    len(
        cell57_target_document_linkage_failures
    )
    == 0
)


target_span_integrity_gate = (
    len(
        cell57_target_span_linkage_failures
    )
    == 0
)


all_target_fields_classified_gate = (
    len(
        cell57_unclassified_target_fields
    )
    == 0

    and

    len(
        cell57_unknown_policy_fields
    )
    == 0
)


v2_target_part_strategy_not_guessed_gate = (
    cell57_v2_target_part_strategy_status
    ==
    "NOT_YET_DECIDED_REQUIRES_EXERCISE_UNIT_TARGET_DESIGN"
)


active_and_staging_registries_unchanged_gate = (
    cell57_target_snapshot_before
    ==
    cell57_target_snapshot_after

    and

    cell57_target_by_case_snapshot_before
    ==
    cell57_target_by_case_snapshot_after

    and

    cell57_target_by_task_snapshot_before
    ==
    cell57_target_by_task_snapshot_after

    and

    cell57_task_snapshot_before
    ==
    cell57_task_snapshot_after

    and

    cell57_golden_snapshot_before
    ==
    cell57_golden_snapshot_after

    and

    cell57_pilot_snapshot_before
    ==
    cell57_pilot_snapshot_after

    and

    cell57_v2_case_staging_snapshot_before
    ==
    cell57_v2_case_staging_snapshot_after

    and

    cell57_v2_task_staging_snapshot_before
    ==
    cell57_v2_task_staging_snapshot_after

    and

    len(
        golden_case_records
    )
    == 724

    and

    len(
        task_draft_plan_rows
    )
    == 724

    and

    len(
        target_evidence_records
    )
    == 572

    and

    len(
        pilot_selection_rows
    )
    == 23
)


v2_target_staging_not_started_gate = (
    EXERCISE_V2_TARGET_CONTRACT_DISCOVERY_V1_DRAFT[
        "v2_target_staging_status"
    ]
    ==
    "NOT_STARTED"
)


case_activation_not_started_gate = (
    EXERCISE_V2_TARGET_CONTRACT_DISCOVERY_V1_DRAFT[
        "case_activation_status"
    ]
    ==
    "NOT_STARTED"
)


task_activation_not_started_gate = (
    EXERCISE_V2_TARGET_CONTRACT_DISCOVERY_V1_DRAFT[
        "task_activation_status"
    ]
    ==
    "NOT_STARTED"
)


pilot_rebuild_not_started_gate = (
    EXERCISE_V2_TARGET_CONTRACT_DISCOVERY_V1_DRAFT[
        "pilot_rebuild_status"
    ]
    ==
    "NOT_STARTED"
)


question_generation_not_started_gate = (
    EXERCISE_V2_TARGET_CONTRACT_DISCOVERY_V1_DRAFT[
        "question_generation_status"
    ]
    ==
    "NOT_STARTED"
)


no_artifact_write_gate = (
    EXERCISE_V2_TARGET_CONTRACT_DISCOVERY_V1_DRAFT[
        "artifact_write_status"
    ]
    ==
    "NONE"
)


cell_57_gates = {
    "exact_38_legacy_exercise_target_records":
        exact_38_targets_gate,

    "legacy_exercise_target_schema_uniform":
        uniform_target_schema_gate,

    "all_38_legacy_case_task_target_links_exact":
        exact_case_task_target_linkage_gate,

    "all_38_targets_resolve_exactly_in_case_index":
        target_case_index_gate,

    "all_38_targets_resolve_exactly_in_task_index":
        target_task_index_gate,

    "exact_target_evidence_id_field_resolved_by_actual_value_shape":
        exactly_one_value_shape_target_id_field_gate,

    "actual_notebook_source_contains_target_id_formula_evidence":
        source_formula_evidence_found_gate,

    "build_target_evidence_id_and_target_part_identity_present":
        actual_target_builder_present_gate,

    "build_target_evidence_id_signature_is_exact":
        actual_target_builder_signature_gate,

    "persisted_target_fields_map_exactly_to_builder_arguments":
        builder_argument_mapping_grounded_gate,

    "actual_target_builder_reproduces_38_of_38":
        actual_builder_reproduces_38_gate,

    "target_ids_independently_recompute_38_of_38":
        independent_target_id_reproduction_gate,

    "target_evidence_id_formula_resolved_without_guessing":
        target_id_formula_resolved_gate,

    "legacy_exercise_targets_each_have_exactly_one_target_part":
        legacy_parts_are_exactly_single_part_gate,

    "legacy_exercise_targets_each_have_exactly_one_text_span":
        legacy_spans_are_exactly_single_span_gate,

    "legacy_target_part_schema_uniform":
        target_part_schema_uniform_gate,

    "target_part_identity_is_nonempty_for_all_38":
        target_part_identity_nonempty_gate,

    "legacy_target_documents_are_subsets_of_case_evidence":
        target_document_linkage_gate,

    "legacy_target_text_spans_and_hashes_are_valid":
        target_span_integrity_gate,

    "every_target_field_has_explicit_v2_handling_policy":
        all_target_fields_classified_gate,

    "v2_target_part_strategy_intentionally_not_guessed_yet":
        v2_target_part_strategy_not_guessed_gate,

    "all_active_and_v2_staging_registries_content_unchanged":
        active_and_staging_registries_unchanged_gate,

    "v2_target_staging_not_started":
        v2_target_staging_not_started_gate,

    "case_activation_not_started":
        case_activation_not_started_gate,

    "task_activation_not_started":
        task_activation_not_started_gate,

    "pilot_rebuild_not_started":
        pilot_rebuild_not_started_gate,

    "question_generation_not_started":
        question_generation_not_started_gate,

    "no_artifact_write":
        no_artifact_write_gate,
}


cell_57_pass = all(
    cell_57_gates.values()
)


# ============================================================
# 24) Output
# ============================================================

print(
    "=== LEGACY EXERCISE TARGET EVIDENCE CONTRACT DISCOVERY — REPAIRED ==="
)


print(
    "Legacy Exercise Targets:",
    len(
        cell57_legacy_target_rows
    )
)


print(
    "Uniform Target schema:",
    cell57_target_schema_uniform
)


print(
    "Target fields:",
    sorted(
        cell57_target_key_union
    )
)


print(
    "\n=== TARGET FIELD DISTINCT COUNTS ==="
)


for field_name, count in sorted(
    cell57_target_field_distinct_counts.items()
):

    print(
        "-",
        field_name,
        ":",
        count,
    )


print(
    "\n=== CONSTANT TARGET FIELDS ==="
)


for field_name, value in sorted(
    cell57_target_constant_fields.items()
):

    print(
        "-",
        field_name,
        "=",
        value,
    )


print(
    "\n=== TARGET ID RESOLUTION ==="
)


print(
    "Value-shape candidates:",
    cell57_target_id_value_shape_candidates
)


print(
    "Resolved Target ID field:",
    cell57_target_id_field_name
)


print(
    "\nValue-shape profile for target_evidence_id:",
    cell57_target_id_value_shape_profiles.get(
        "target_evidence_id"
    )
)


print(
    "Builder:",
    "build_target_evidence_id"
    if cell57_actual_builder_present
    else None
)


print(
    "Builder signature:",
    cell57_builder_signature
)


print(
    "Persisted -> builder arg mapping:",
    CELL57_TARGET_BUILDER_ARGUMENT_MAPPING
)


print(
    "Builder reproduction matches:",
    cell57_builder_exact_match_count,
    "/ 38"
)


print(
    "Builder failures:",
    len(
        cell57_builder_reproduction_failures
    )
)


if (
    cell57_builder_reproduction_failures
):

    print(
        "First builder failures:"
    )

    for row in (
        cell57_builder_reproduction_failures[
            :10
        ]
    ):

        print(
            "-",
            row
        )


print(
    "Independent ID failures:",
    len(
        cell57_independent_target_id_failures
    )
)


print(
    "Formula status:",
    cell57_target_id_formula_status
)


print(
    "\n=== LEGACY TARGET SHAPE ==="
)


print(
    "Target parts per case:",
    dict(
        cell57_target_parts_length_distribution
    )
)


print(
    "Text spans per case:",
    dict(
        cell57_target_text_spans_length_distribution
    )
)


print(
    "Target documents per case:",
    dict(
        cell57_target_document_count_distribution
    )
)


print(
    "Target Part fields:",
    sorted(
        cell57_target_part_key_union
    )
)


print(
    "Target Part schema uniform:",
    cell57_target_part_schema_uniform
)


print(
    "Target Part identity failures:",
    len(
        cell57_part_identity_recompute_failures
    )
)


print(
    "Target document linkage failures:",
    len(
        cell57_target_document_linkage_failures
    )
)


print(
    "Target span integrity failures:",
    len(
        cell57_target_span_linkage_failures
    )
)


print(
    "\n=== V2 TARGET FIELD POLICY ==="
)


for field_name, policy in sorted(
    cell57_v2_target_field_policy.items()
):

    print(
        "-",
        field_name,
        "=>",
        policy,
    )


print(
    "\nUnclassified Target fields:",
    cell57_unclassified_target_fields
)


print(
    "Unknown policy fields:",
    cell57_unknown_policy_fields
)


print(
    "\nV2 Target Part strategy:",
    cell57_v2_target_part_strategy_status
)


print(
    "\n=== ACTIVE COUNTS ==="
)


print(
    "Golden:",
    len(
        golden_case_records
    )
)


print(
    "Tasks:",
    len(
        task_draft_plan_rows
    )
)


print(
    "Targets:",
    len(
        target_evidence_records
    )
)


print(
    "Pilots:",
    len(
        pilot_selection_rows
    )
)


print(
    "\n=== CODE CELL 57 GATES ==="
)


for name, passed in (
    cell_57_gates.items()
):

    print(
        name,
        ":",
        "PASS"
        if passed
        else "FAIL"
    )


print(
    "\nCODE CELL 57 GATE:",
    "PASS"
    if cell_57_pass
    else "FAIL"
)


assert cell_57_pass, (
    "Exercise Target Evidence contract discovery remains "
    "invalid. Do not stage V2 Targets. Repair this SAME "
    "Code Cell 57 until target_evidence_id is reproduced "
    "38/38 both by the actual builder and independent formula "
    "recomputation, with valid Target Part and span integrity."
)
```

```python
# Code Cell 58 — REPAIRED AGAIN (NO math / statistics dependency)

# ============================================================
# Exercise-Unit V2 Target Design Feasibility Audit
#
# REPAIR:
#   - no `statistics`
#   - no `math`
#   - median and nearest-rank percentiles are computed locally
#
# Candidate strategy:
#   FULL_EXERCISE_UNIT_ORDERED_DOCUMENT_PARTS
#
# DESIGN / FEASIBILITY ONLY.
#
# NO V2 Target materialization.
# NO active registry mutation.
# NO Case activation.
# NO Task activation.
# NO Pilot rebuild.
# NO question generation.
# NO artifact writes.
# ============================================================


# ============================================================
# 1) Snapshot active and existing staging state
# ============================================================

cell58_target_snapshot_before = hashlib.sha256(
    stable_json_key(
        target_evidence_records
    ).encode("utf-8")
).hexdigest()


cell58_task_snapshot_before = hashlib.sha256(
    stable_json_key(
        task_draft_plan_rows
    ).encode("utf-8")
).hexdigest()


cell58_golden_snapshot_before = hashlib.sha256(
    stable_json_key(
        golden_case_records
    ).encode("utf-8")
).hexdigest()


cell58_pilot_snapshot_before = hashlib.sha256(
    stable_json_key(
        pilot_selection_rows
    ).encode("utf-8")
).hexdigest()


cell58_case_staging_snapshot_before = hashlib.sha256(
    stable_json_key(
        staged_exercise_v2_golden_case_records
    ).encode("utf-8")
).hexdigest()


cell58_task_staging_snapshot_before = hashlib.sha256(
    stable_json_key(
        staged_exercise_v2_task_draft_plan_rows
    ).encode("utf-8")
).hexdigest()


# ============================================================
# 2) Existing Target system multi-part capability
# ============================================================

cell58_active_target_part_count_distribution = Counter(
    len(
        row[
            "target_parts"
        ]
    )
    for row
    in target_evidence_records
)


cell58_active_target_max_part_count = max(
    cell58_active_target_part_count_distribution
)


cell58_existing_multi_part_target_count = sum(
    count
    for part_count, count
    in cell58_active_target_part_count_distribution.items()
    if part_count > 1
)


# ============================================================
# 3) Legacy Exercise Target Part baseline
# ============================================================

cell58_legacy_target_parts = [
    part
    for target_row
    in cell57_legacy_target_rows
    for part
    in target_row[
        "target_parts"
    ]
]


cell58_legacy_target_part_role_counts = Counter(
    part.get(
        "role"
    )
    for part
    in cell58_legacy_target_parts
)


cell58_legacy_target_part_speaker_state_counts = Counter(
    stable_json_key(
        part.get(
            "speaker_state"
        )
    )
    for part
    in cell58_legacy_target_parts
)


cell58_legacy_part_schema = set(
    cell57_target_part_key_union
)


# ============================================================
# 4) Verify whether legacy Exercise Parts are full-document
# ============================================================

cell58_legacy_full_document_span_failures = []


for target_row in (
    cell57_legacy_target_rows
):

    for part in (
        target_row[
            "target_parts"
        ]
    ):

        document_id = (
            part[
                "document_id"
            ]
        )


        if (
            document_id
            not in
            corpus_target_index
        ):

            cell58_legacy_full_document_span_failures.append(
                {
                    "document_id":
                        document_id,

                    "reason":
                        "DOCUMENT_NOT_IN_CORPUS_TARGET_INDEX",
                }
            )

            continue


        document_text = (
            corpus_target_index[
                document_id
            ][
                "text"
            ]
        )


        if not (
            part[
                "char_start"
            ]
            ==
            0

            and

            part[
                "char_end"
            ]
            ==
            len(
                document_text
            )
        ):

            cell58_legacy_full_document_span_failures.append(
                {
                    "document_id":
                        document_id,

                    "reason":
                        "LEGACY_EXERCISE_PART_NOT_FULL_DOCUMENT",

                    "char_start":
                        part[
                            "char_start"
                        ],

                    "char_end":
                        part[
                            "char_end"
                        ],

                    "document_length":
                        len(
                            document_text
                        ),
                }
            )


cell58_legacy_exercise_parts_are_full_document = (
    len(
        cell58_legacy_full_document_span_failures
    )
    ==
    0
)


# ============================================================
# 5) Discover actual target_part_from_document callable
# ============================================================

cell58_target_part_builder = globals().get(
    "target_part_from_document"
)


cell58_target_part_builder_present = callable(
    cell58_target_part_builder
)


cell58_target_part_builder_signature = []

cell58_target_part_builder_defaults = ()


if (
    cell58_target_part_builder_present
):

    cell58_builder_code = (
        cell58_target_part_builder.__code__
    )


    cell58_target_part_builder_signature = list(
        cell58_builder_code.co_varnames[
            :cell58_builder_code.co_argcount
        ]
    )


    cell58_target_part_builder_defaults = (
        cell58_target_part_builder.__defaults__
        or ()
    )


cell58_target_part_builder_arg_count = len(
    cell58_target_part_builder_signature
)


cell58_target_part_builder_default_count = len(
    cell58_target_part_builder_defaults
)


cell58_target_part_builder_required_count = (
    cell58_target_part_builder_arg_count
    -
    cell58_target_part_builder_default_count
)


cell58_target_part_builder_required_args = (
    cell58_target_part_builder_signature[
        :cell58_target_part_builder_required_count
    ]
)


cell58_target_part_builder_optional_args = (
    cell58_target_part_builder_signature[
        cell58_target_part_builder_required_count:
    ]
)


# ============================================================
# 6) Source inspection
# ============================================================

cell58_target_part_source_snippets = []


for code_cell in (
    cell49_code_cells
):

    source_lines = (
        code_cell[
            "source"
        ].splitlines()
    )


    for line_index, line in enumerate(
        source_lines
    ):

        if (
            "def target_part_from_document"
            not in
            line
        ):

            continue


        start_index = max(
            0,
            line_index - 5,
        )


        end_index = min(
            len(
                source_lines
            ),
            line_index + 90,
        )


        cell58_target_part_source_snippets.append(
            {
                "cell_index":
                    code_cell[
                        "cell_index"
                    ],

                "line_index":
                    line_index,

                "snippet":
                    "\n".join(
                        source_lines[
                            start_index:
                            end_index
                        ]
                    ),
            }
        )


# ============================================================
# 7) Resolve uniform Legacy Exercise role
# ============================================================

cell58_uniform_legacy_role = None


if (
    len(
        cell58_legacy_target_part_role_counts
    )
    ==
    1
):

    cell58_uniform_legacy_role = next(
        iter(
            cell58_legacy_target_part_role_counts.keys()
        )
    )


# ============================================================
# 8) Ground required builder arguments
# ============================================================

CELL58_SUPPORTED_PART_BUILDER_ARGUMENTS = {
    "document_id",
    "role",
    "char_start",
    "char_end",
}


cell58_unknown_required_builder_args = sorted(
    set(
        cell58_target_part_builder_required_args
    )
    -
    CELL58_SUPPORTED_PART_BUILDER_ARGUMENTS
)


cell58_builder_role_required = (
    "role"
    in
    cell58_target_part_builder_required_args
)


cell58_builder_span_args_required = bool(
    {
        "char_start",
        "char_end",
    }
    &
    set(
        cell58_target_part_builder_required_args
    )
)


cell58_builder_arguments_groundable = (
    cell58_target_part_builder_present

    and

    len(
        cell58_unknown_required_builder_args
    )
    ==
    0

    and

    (
        not
        cell58_builder_role_required

        or

        cell58_uniform_legacy_role
        is not None
    )

    and

    (
        not
        cell58_builder_span_args_required

        or

        cell58_legacy_exercise_parts_are_full_document
    )
)


# ============================================================
# 9) Full selected V2 Evidence universe
# ============================================================

cell58_selected_document_occurrences = [
    document_id

    for staged_case
    in staged_exercise_v2_golden_case_records

    for document_id
    in staged_case[
        "evidence_document_ids"
    ]
]


cell58_unique_selected_document_ids = set(
    cell58_selected_document_occurrences
)


cell58_total_selected_document_occurrences = len(
    cell58_selected_document_occurrences
)


cell58_unique_selected_document_count = len(
    cell58_unique_selected_document_ids
)


cell58_cross_case_duplicate_occurrences = (
    cell58_total_selected_document_occurrences
    -
    cell58_unique_selected_document_count
)


# ============================================================
# 10) Safe real-document Part builder
# ============================================================

def cell58_build_document_part(
    document_id,
):

    if not (
        cell58_builder_arguments_groundable
    ):

        raise AssertionError(
            "target_part_from_document required arguments "
            "cannot be grounded safely."
        )


    if (
        document_id
        not in
        corpus_target_index
    ):

        raise AssertionError(
            "Document missing from corpus_target_index: "
            f"{document_id}"
        )


    document_text = (
        corpus_target_index[
            document_id
        ][
            "text"
        ]
    )


    kwargs = {}


    for arg_name in (
        cell58_target_part_builder_required_args
    ):

        if (
            arg_name
            ==
            "document_id"
        ):

            kwargs[
                arg_name
            ] = document_id


        elif (
            arg_name
            ==
            "role"
        ):

            kwargs[
                arg_name
            ] = (
                cell58_uniform_legacy_role
            )


        elif (
            arg_name
            ==
            "char_start"
        ):

            kwargs[
                arg_name
            ] = 0


        elif (
            arg_name
            ==
            "char_end"
        ):

            kwargs[
                arg_name
            ] = len(
                document_text
            )


        else:

            raise AssertionError(
                "Unsupported required builder argument: "
                f"{arg_name}"
            )


    return (
        cell58_target_part_builder(
            **kwargs
        )
    )


# ============================================================
# 11) Preflight all unique selected Evidence documents
# ============================================================

cell58_document_part_preflight = {}

cell58_document_part_build_failures = []


for document_id in sorted(
    cell58_unique_selected_document_ids
):

    try:

        part = (
            cell58_build_document_part(
                document_id
            )
        )


    except Exception as exc:

        cell58_document_part_build_failures.append(
            {
                "document_id":
                    document_id,

                "error_type":
                    type(
                        exc
                    ).__name__,

                "error_message":
                    str(
                        exc
                    ),
            }
        )

        continue


    cell58_document_part_preflight[
        document_id
    ] = part


# ============================================================
# 12) Part schema audit
# ============================================================

cell58_part_schema_failures = []


for document_id, part in (
    cell58_document_part_preflight.items()
):

    if not isinstance(
        part,
        dict,
    ):

        cell58_part_schema_failures.append(
            {
                "document_id":
                    document_id,

                "reason":
                    "PART_NOT_DICT",

                "actual_type":
                    type(
                        part
                    ).__name__,
            }
        )

        continue


    if (
        set(
            part.keys()
        )
        !=
        cell58_legacy_part_schema
    ):

        cell58_part_schema_failures.append(
            {
                "document_id":
                    document_id,

                "reason":
                    "PART_SCHEMA_MISMATCH",

                "actual_fields":
                    sorted(
                        part.keys()
                    ),

                "legacy_fields":
                    sorted(
                        cell58_legacy_part_schema
                    ),
            }
        )


# ============================================================
# 13) Independent Part integrity audit
# ============================================================

cell58_part_integrity_failures = []


for document_id, part in (
    cell58_document_part_preflight.items()
):

    reasons = []


    if (
        document_id
        not in
        corpus_target_index
    ):

        reasons.append(
            "DOCUMENT_MISSING_FROM_CORPUS_TARGET_INDEX"
        )


    else:

        corpus_record = (
            corpus_target_index[
                document_id
            ]
        )


        text = (
            corpus_record[
                "text"
            ]
        )


        if (
            part.get(
                "document_id"
            )
            !=
            document_id
        ):

            reasons.append(
                "DOCUMENT_ID_MISMATCH"
            )


        char_start = (
            part.get(
                "char_start"
            )
        )


        char_end = (
            part.get(
                "char_end"
            )
        )


        valid_bounds = (
            isinstance(
                char_start,
                int,
            )

            and

            isinstance(
                char_end,
                int,
            )

            and

            0
            <=
            char_start
            <
            char_end
            <=
            len(
                text
            )
        )


        if not (
            valid_bounds
        ):

            reasons.append(
                "INVALID_CHAR_BOUNDS"
            )


        else:

            part_text = (
                text[
                    char_start:
                    char_end
                ]
            )


            if not (
                part_text.strip()
            ):

                reasons.append(
                    "EMPTY_TARGET_TEXT"
                )


            if (
                part.get(
                    "char_count"
                )
                !=
                len(
                    part_text
                )
            ):

                reasons.append(
                    "CHAR_COUNT_MISMATCH"
                )


            recomputed_sha = hashlib.sha256(
                part_text.encode(
                    "utf-8"
                )
            ).hexdigest()


            if (
                recomputed_sha
                !=
                part.get(
                    "target_text_sha256"
                )
            ):

                reasons.append(
                    "TARGET_TEXT_SHA_MISMATCH"
                )


        if (
            part.get(
                "source_id"
            )
            !=
            corpus_record.get(
                "source_id"
            )
        ):

            reasons.append(
                "SOURCE_ID_MISMATCH"
            )


        if (
            stable_json_key(
                part.get(
                    "locator"
                )
            )
            !=
            stable_json_key(
                corpus_record.get(
                    "locator"
                )
            )
        ):

            reasons.append(
                "LOCATOR_MISMATCH"
            )


        if (
            stable_json_key(
                part.get(
                    "attribution"
                )
            )
            !=
            stable_json_key(
                corpus_record.get(
                    "attribution"
                )
            )
        ):

            reasons.append(
                "ATTRIBUTION_MISMATCH"
            )


    if (
        reasons
    ):

        cell58_part_integrity_failures.append(
            {
                "document_id":
                    document_id,

                "reasons":
                    reasons,
            }
        )


# ============================================================
# 14) Build 38 FULL-UNIT candidate Part sets
# ============================================================

cell58_full_unit_candidate_rows = []

cell58_full_unit_build_failures = []


for staged_case in (
    staged_exercise_v2_golden_case_records
):

    case_id = (
        staged_case[
            "golden_case_id"
        ]
    )


    document_ids = list(
        staged_case[
            "evidence_document_ids"
        ]
    )


    missing_part_document_ids = [
        document_id

        for document_id
        in document_ids

        if (
            document_id
            not in
            cell58_document_part_preflight
        )
    ]


    if (
        missing_part_document_ids
    ):

        cell58_full_unit_build_failures.append(
            {
                "golden_case_id":
                    case_id,

                "reason":
                    "DOCUMENT_PARTS_MISSING",

                "missing_document_ids":
                    missing_part_document_ids,
            }
        )

        continue


    candidate_parts = [
        cell52_safe_clone(
            cell58_document_part_preflight[
                document_id
            ]
        )

        for document_id
        in document_ids
    ]


    cell58_full_unit_candidate_rows.append(
        {
            "golden_case_id":
                case_id,

            "population_unit_id":
                staged_case[
                    "candidate_identity"
                ][
                    "population_unit_id"
                ],

            "heading_path":
                cell52_safe_clone(
                    staged_case[
                        "candidate_snapshot"
                    ][
                        "heading_path"
                    ]
                ),

            "evidence_document_ids":
                document_ids,

            "candidate_parts":
                candidate_parts,

            "candidate_part_count":
                len(
                    candidate_parts
                ),

            "candidate_total_tokens":
                sum(
                    part[
                        "token_count"
                    ]
                    for part
                    in candidate_parts
                ),

            "candidate_total_chars":
                sum(
                    part[
                        "char_count"
                    ]
                    for part
                    in candidate_parts
                ),
        }
    )


# ============================================================
# 15) Exact coverage and ordering audit
# ============================================================

cell58_full_unit_integrity_failures = []


for row in (
    cell58_full_unit_candidate_rows
):

    part_document_ids = [
        part[
            "document_id"
        ]
        for part
        in row[
            "candidate_parts"
        ]
    ]


    reasons = []


    if (
        part_document_ids
        !=
        row[
            "evidence_document_ids"
        ]
    ):

        reasons.append(
            "PART_ORDER_MISMATCH"
        )


    if (
        len(
            part_document_ids
        )
        !=
        len(
            set(
                part_document_ids
            )
        )
    ):

        reasons.append(
            "DUPLICATE_DOCUMENT_WITHIN_EXERCISE_UNIT"
        )


    if (
        len(
            part_document_ids
        )
        !=
        row[
            "candidate_part_count"
        ]
    ):

        reasons.append(
            "PART_COUNT_MISMATCH"
        )


    if (
        reasons
    ):

        cell58_full_unit_integrity_failures.append(
            {
                "golden_case_id":
                    row[
                        "golden_case_id"
                    ],

                "reasons":
                    reasons,
            }
        )


# ============================================================
# 16) Staged V2 Task lookup
# ============================================================

cell58_staged_task_by_case_id = {
    row[
        "golden_case_id"
    ]:
        row

    for row
    in staged_exercise_v2_task_draft_plan_rows
}


# ============================================================
# 17) Build candidate multi-part Target IDs
# ============================================================

cell58_candidate_target_id_rows = []

cell58_candidate_target_id_failures = []


for row in (
    cell58_full_unit_candidate_rows
):

    case_id = (
        row[
            "golden_case_id"
        ]
    )


    if (
        case_id
        not in
        cell58_staged_task_by_case_id
    ):

        cell58_candidate_target_id_failures.append(
            {
                "golden_case_id":
                    case_id,

                "reason":
                    "STAGED_TASK_MISSING",
            }
        )

        continue


    task_row = (
        cell58_staged_task_by_case_id[
            case_id
        ]
    )


    try:

        candidate_target_id = (
            build_target_evidence_id(
                task_draft_id=(
                    task_row[
                        "task_draft_id"
                    ]
                ),
                method=(
                    EXERCISE_V2_TASK_EVIDENCE_TARGET_METHOD
                ),
                parts=(
                    row[
                        "candidate_parts"
                    ]
                ),
            )
        )


    except Exception as exc:

        cell58_candidate_target_id_failures.append(
            {
                "golden_case_id":
                    case_id,

                "reason":
                    "TARGET_ID_BUILD_EXCEPTION",

                "error_type":
                    type(
                        exc
                    ).__name__,

                "error_message":
                    str(
                        exc
                    ),
            }
        )

        continue


    cell58_candidate_target_id_rows.append(
        {
            "golden_case_id":
                case_id,

            "task_draft_id":
                task_row[
                    "task_draft_id"
                ],

            "candidate_target_evidence_id":
                candidate_target_id,

            "part_count":
                row[
                    "candidate_part_count"
                ],
        }
    )


# ============================================================
# 18) Independent Target-ID recomputation
# ============================================================

cell58_candidate_target_id_by_case = {
    row[
        "golden_case_id"
    ]:
        row[
            "candidate_target_evidence_id"
        ]

    for row
    in cell58_candidate_target_id_rows
}


cell58_candidate_target_id_recompute_failures = []


for candidate_row in (
    cell58_full_unit_candidate_rows
):

    case_id = (
        candidate_row[
            "golden_case_id"
        ]
    )


    if (
        case_id
        not in
        cell58_candidate_target_id_by_case
    ):

        continue


    task_row = (
        cell58_staged_task_by_case_id[
            case_id
        ]
    )


    independent_payload = {
        "schema_version":
            TARGET_EVIDENCE_SCHEMA_VERSION,

        "task_draft_id":
            task_row[
                "task_draft_id"
            ],

        "method":
            EXERCISE_V2_TASK_EVIDENCE_TARGET_METHOD,

        "parts":
            [
                target_part_identity(
                    part
                )
                for part
                in candidate_row[
                    "candidate_parts"
                ]
            ],
    }


    digest = hashlib.sha256(
        stable_json_key(
            independent_payload
        ).encode(
            "utf-8"
        )
    ).hexdigest()


    independent_target_id = (
        "target-evidence:"
        +
        digest[
            :32
        ]
    )


    if (
        independent_target_id
        !=
        cell58_candidate_target_id_by_case[
            case_id
        ]
    ):

        cell58_candidate_target_id_recompute_failures.append(
            {
                "golden_case_id":
                    case_id,

                "builder_id":
                    cell58_candidate_target_id_by_case[
                        case_id
                    ],

                "independent_id":
                    independent_target_id,
            }
        )


# ============================================================
# 19) Candidate ID uniqueness / collision
# ============================================================

cell58_candidate_target_ids = [
    row[
        "candidate_target_evidence_id"
    ]
    for row
    in cell58_candidate_target_id_rows
]


cell58_candidate_target_id_counts = Counter(
    cell58_candidate_target_ids
)


cell58_duplicate_candidate_target_ids = sorted(
    [
        target_id

        for target_id, count
        in cell58_candidate_target_id_counts.items()

        if count > 1
    ]
)


cell58_active_target_ids = {
    row[
        "target_evidence_id"
    ]
    for row
    in target_evidence_records
}


cell58_candidate_target_id_collisions_with_active = sorted(
    set(
        cell58_candidate_target_ids
    )
    &
    cell58_active_target_ids
)


# ============================================================
# 20) Part-count and token distributions
#
# No math/statistics imports.
# ============================================================

cell58_candidate_part_count_distribution = Counter(
    row[
        "candidate_part_count"
    ]
    for row
    in cell58_full_unit_candidate_rows
)


cell58_candidate_token_values = sorted(
    row[
        "candidate_total_tokens"
    ]
    for row
    in cell58_full_unit_candidate_rows
)


cell58_candidate_char_values = sorted(
    row[
        "candidate_total_chars"
    ]
    for row
    in cell58_full_unit_candidate_rows
)


def cell58_median(
    values,
):

    if not values:

        return None


    ordered_values = sorted(
        values
    )


    count = len(
        ordered_values
    )


    midpoint = (
        count
        //
        2
    )


    if (
        count
        %
        2
        ==
        1
    ):

        return ordered_values[
            midpoint
        ]


    return (
        ordered_values[
            midpoint - 1
        ]
        +
        ordered_values[
            midpoint
        ]
    ) / 2


def cell58_nearest_rank_percentile(
    values,
    numerator,
    denominator,
):

    if not values:

        return None


    if not (
        isinstance(
            numerator,
            int,
        )

        and

        isinstance(
            denominator,
            int,
        )

        and

        denominator > 0

        and

        0 < numerator <= denominator
    ):

        raise ValueError(
            "Percentile numerator/denominator invalid."
        )


    # ceil(n * numerator / denominator)
    rank = (
        (
            len(
                values
            )
            *
            numerator
        )
        +
        denominator
        -
        1
    ) // denominator


    rank = max(
        1,
        min(
            rank,
            len(
                values
            ),
        ),
    )


    return values[
        rank - 1
    ]


cell58_token_summary = {
    "min":
        min(
            cell58_candidate_token_values
        )
        if cell58_candidate_token_values
        else None,

    "median":
        cell58_median(
            cell58_candidate_token_values
        ),

    "p75":
        cell58_nearest_rank_percentile(
            cell58_candidate_token_values,
            75,
            100,
        ),

    "p90":
        cell58_nearest_rank_percentile(
            cell58_candidate_token_values,
            90,
            100,
        ),

    "max":
        max(
            cell58_candidate_token_values
        )
        if cell58_candidate_token_values
        else None,
}


cell58_largest_candidate_rows = sorted(
    cell58_full_unit_candidate_rows,
    key=lambda row:
        (
            row[
                "candidate_total_tokens"
            ],
            row[
                "candidate_part_count"
            ],
        ),
    reverse=True,
)


# ============================================================
# 21) Boundary leakage audit
# ============================================================

cell58_boundary_leakage_failures = []


cell58_case_by_id = {
    row[
        "golden_case_id"
    ]:
        row

    for row
    in staged_exercise_v2_golden_case_records
}


for candidate_row in (
    cell58_full_unit_candidate_rows
):

    case_id = (
        candidate_row[
            "golden_case_id"
        ]
    )


    expected_document_ids = list(
        cell58_case_by_id[
            case_id
        ][
            "evidence_document_ids"
        ]
    )


    actual_document_ids = [
        part[
            "document_id"
        ]
        for part
        in candidate_row[
            "candidate_parts"
        ]
    ]


    if (
        actual_document_ids
        !=
        expected_document_ids
    ):

        cell58_boundary_leakage_failures.append(
            {
                "golden_case_id":
                    case_id,

                "expected_document_ids":
                    expected_document_ids,

                "actual_document_ids":
                    actual_document_ids,
            }
        )


# ============================================================
# 22) Metadata diversity profiles
# ============================================================

cell58_unit_metadata_profiles = []


for row in (
    cell58_full_unit_candidate_rows
):

    parts = (
        row[
            "candidate_parts"
        ]
    )


    cell58_unit_metadata_profiles.append(
        {
            "golden_case_id":
                row[
                    "golden_case_id"
                ],

            "part_count":
                row[
                    "candidate_part_count"
                ],

            "source_id_count":
                len(
                    {
                        part[
                            "source_id"
                        ]
                        for part
                        in parts
                    }
                ),

            "locator_count":
                len(
                    {
                        stable_json_key(
                            part[
                                "locator"
                            ]
                        )
                        for part
                        in parts
                    }
                ),

            "attribution_state_count":
                len(
                    {
                        stable_json_key(
                            part[
                                "attribution"
                            ]
                        )
                        for part
                        in parts
                    }
                ),

            "speaker_state_count":
                len(
                    {
                        stable_json_key(
                            part[
                                "speaker_state"
                            ]
                        )
                        for part
                        in parts
                    }
                ),
        }
    )


# ============================================================
# 23) Technical feasibility
# ============================================================

cell58_full_unit_strategy_technical_feasibility = (
    cell58_existing_multi_part_target_count
    >
    0

    and

    cell58_builder_arguments_groundable

    and

    len(
        cell58_document_part_build_failures
    )
    ==
    0

    and

    len(
        cell58_part_schema_failures
    )
    ==
    0

    and

    len(
        cell58_part_integrity_failures
    )
    ==
    0

    and

    len(
        cell58_full_unit_build_failures
    )
    ==
    0

    and

    len(
        cell58_full_unit_integrity_failures
    )
    ==
    0

    and

    len(
        cell58_candidate_target_id_failures
    )
    ==
    0

    and

    len(
        cell58_candidate_target_id_recompute_failures
    )
    ==
    0

    and

    len(
        cell58_duplicate_candidate_target_ids
    )
    ==
    0

    and

    len(
        cell58_candidate_target_id_collisions_with_active
    )
    ==
    0

    and

    len(
        cell58_boundary_leakage_failures
    )
    ==
    0
)


cell58_full_unit_strategy_benchmark_status = (
    "TECHNICALLY_FEASIBLE_PENDING_CONTEXT_AND_EVIDENCE_COMPLETENESS_DECISION"
    if cell58_full_unit_strategy_technical_feasibility
    else
    "TECHNICALLY_NOT_FEASIBLE"
)


# ============================================================
# 24) Candidate design registry
# ============================================================

EXERCISE_V2_TARGET_DESIGN_FEASIBILITY_V1_DRAFT = {
    "version":
        "exercise_v2_target_design_feasibility_v1_draft",

    "candidate_strategy":
        "FULL_EXERCISE_UNIT_ORDERED_DOCUMENT_PARTS",

    "candidate_semantics": {
        "one_real_part_per_document":
            True,

        "exact_case_order":
            True,

        "synthetic_aggregation":
            False,

        "semantic_subselection":
            False,

        "cross_boundary_evidence":
            False,
    },

    "legacy_exercise_parts_are_full_document":
        cell58_legacy_exercise_parts_are_full_document,

    "existing_active_multi_part_targets":
        cell58_existing_multi_part_target_count,

    "existing_active_max_part_count":
        cell58_active_target_max_part_count,

    "selected_case_count":
        len(
            cell58_full_unit_candidate_rows
        ),

    "selected_document_occurrence_count":
        cell58_total_selected_document_occurrences,

    "unique_selected_document_count":
        cell58_unique_selected_document_count,

    "technical_feasibility":
        cell58_full_unit_strategy_technical_feasibility,

    "benchmark_status":
        cell58_full_unit_strategy_benchmark_status,

    "target_staging_status":
        "NOT_STARTED",

    "case_activation_status":
        "NOT_STARTED",

    "task_activation_status":
        "NOT_STARTED",

    "pilot_rebuild_status":
        "NOT_STARTED",

    "question_generation_status":
        "NOT_STARTED",

    "artifact_write_status":
        "NONE",
}


# ============================================================
# 25) No-mutation recheck
# ============================================================

cell58_target_snapshot_after = hashlib.sha256(
    stable_json_key(
        target_evidence_records
    ).encode("utf-8")
).hexdigest()


cell58_task_snapshot_after = hashlib.sha256(
    stable_json_key(
        task_draft_plan_rows
    ).encode("utf-8")
).hexdigest()


cell58_golden_snapshot_after = hashlib.sha256(
    stable_json_key(
        golden_case_records
    ).encode("utf-8")
).hexdigest()


cell58_pilot_snapshot_after = hashlib.sha256(
    stable_json_key(
        pilot_selection_rows
    ).encode("utf-8")
).hexdigest()


cell58_case_staging_snapshot_after = hashlib.sha256(
    stable_json_key(
        staged_exercise_v2_golden_case_records
    ).encode("utf-8")
).hexdigest()


cell58_task_staging_snapshot_after = hashlib.sha256(
    stable_json_key(
        staged_exercise_v2_task_draft_plan_rows
    ).encode("utf-8")
).hexdigest()


# ============================================================
# 26) Gates
# ============================================================

cell58_gates = {
    "legacy_exercise_target_parts_exactly_38":
        (
            len(
                cell58_legacy_target_parts
            )
            ==
            38
        ),

    "legacy_exercise_part_role_uniform":
        (
            len(
                cell58_legacy_target_part_role_counts
            )
            ==
            1

            and

            cell58_uniform_legacy_role
            is not None
        ),

    "legacy_exercise_parts_are_empirically_full_document_spans":
        cell58_legacy_exercise_parts_are_full_document,

    "existing_target_contract_already_supports_multi_part_targets":
        (
            cell58_existing_multi_part_target_count
            >
            0

            and

            cell58_active_target_max_part_count
            >
            1
        ),

    "actual_target_part_from_document_builder_present":
        cell58_target_part_builder_present,

    "target_part_builder_source_found":
        (
            len(
                cell58_target_part_source_snippets
            )
            >
            0
        ),

    "all_required_target_part_builder_arguments_are_grounded":
        cell58_builder_arguments_groundable,

    "no_unknown_required_target_part_builder_arguments":
        (
            len(
                cell58_unknown_required_builder_args
            )
            ==
            0
        ),

    "selected_v2_evidence_occurrence_count_is_480":
        (
            cell58_total_selected_document_occurrences
            ==
            480
        ),

    "selected_v2_evidence_unique_document_count_is_480":
        (
            cell58_unique_selected_document_count
            ==
            480
        ),

    "all_480_real_document_parts_build_successfully":
        (
            len(
                cell58_document_part_preflight
            )
            ==
            480

            and

            len(
                cell58_document_part_build_failures
            )
            ==
            0
        ),

    "all_480_parts_match_existing_target_part_schema":
        (
            len(
                cell58_part_schema_failures
            )
            ==
            0
        ),

    "all_480_parts_pass_independent_integrity_checks":
        (
            len(
                cell58_part_integrity_failures
            )
            ==
            0
        ),

    "exact_38_full_unit_candidate_part_sets_constructed":
        (
            len(
                cell58_full_unit_candidate_rows
            )
            ==
            38

            and

            len(
                cell58_full_unit_build_failures
            )
            ==
            0
        ),

    "all_38_candidate_part_sets_preserve_exact_case_order_and_coverage":
        (
            len(
                cell58_full_unit_integrity_failures
            )
            ==
            0
        ),

    "exact_38_candidate_multi_part_target_ids_build":
        (
            len(
                cell58_candidate_target_id_rows
            )
            ==
            38

            and

            len(
                cell58_candidate_target_id_failures
            )
            ==
            0
        ),

    "all_38_candidate_target_ids_independently_recompute":
        (
            len(
                cell58_candidate_target_id_recompute_failures
            )
            ==
            0
        ),

    "all_38_candidate_target_ids_unique":
        (
            len(
                cell58_candidate_target_ids
            )
            ==
            38

            and

            len(
                set(
                    cell58_candidate_target_ids
                )
            )
            ==
            38

            and

            len(
                cell58_duplicate_candidate_target_ids
            )
            ==
            0
        ),

    "no_candidate_target_id_collision_with_active_targets":
        (
            len(
                cell58_candidate_target_id_collisions_with_active
            )
            ==
            0
        ),

    "no_candidate_target_crosses_adjudicated_exercise_boundary":
        (
            len(
                cell58_boundary_leakage_failures
            )
            ==
            0
        ),

    "full_exercise_unit_multi_part_target_is_technically_representable":
        cell58_full_unit_strategy_technical_feasibility,

    "benchmark_suitability_intentionally_remains_open":
        (
            cell58_full_unit_strategy_benchmark_status
            ==
            "TECHNICALLY_FEASIBLE_PENDING_CONTEXT_AND_EVIDENCE_COMPLETENESS_DECISION"
        ),

    "all_active_and_existing_staging_registries_unchanged":
        (
            cell58_target_snapshot_before
            ==
            cell58_target_snapshot_after

            and

            cell58_task_snapshot_before
            ==
            cell58_task_snapshot_after

            and

            cell58_golden_snapshot_before
            ==
            cell58_golden_snapshot_after

            and

            cell58_pilot_snapshot_before
            ==
            cell58_pilot_snapshot_after

            and

            cell58_case_staging_snapshot_before
            ==
            cell58_case_staging_snapshot_after

            and

            cell58_task_staging_snapshot_before
            ==
            cell58_task_staging_snapshot_after

            and

            len(
                golden_case_records
            )
            ==
            724

            and

            len(
                task_draft_plan_rows
            )
            ==
            724

            and

            len(
                target_evidence_records
            )
            ==
            572

            and

            len(
                pilot_selection_rows
            )
            ==
            23
        ),

    "v2_target_staging_not_started":
        (
            EXERCISE_V2_TARGET_DESIGN_FEASIBILITY_V1_DRAFT[
                "target_staging_status"
            ]
            ==
            "NOT_STARTED"
        ),

    "case_activation_not_started":
        (
            EXERCISE_V2_TARGET_DESIGN_FEASIBILITY_V1_DRAFT[
                "case_activation_status"
            ]
            ==
            "NOT_STARTED"
        ),

    "task_activation_not_started":
        (
            EXERCISE_V2_TARGET_DESIGN_FEASIBILITY_V1_DRAFT[
                "task_activation_status"
            ]
            ==
            "NOT_STARTED"
        ),

    "pilot_rebuild_not_started":
        (
            EXERCISE_V2_TARGET_DESIGN_FEASIBILITY_V1_DRAFT[
                "pilot_rebuild_status"
            ]
            ==
            "NOT_STARTED"
        ),

    "question_generation_not_started":
        (
            EXERCISE_V2_TARGET_DESIGN_FEASIBILITY_V1_DRAFT[
                "question_generation_status"
            ]
            ==
            "NOT_STARTED"
        ),

    "no_artifact_write":
        (
            EXERCISE_V2_TARGET_DESIGN_FEASIBILITY_V1_DRAFT[
                "artifact_write_status"
            ]
            ==
            "NONE"
        ),
}


cell_58_pass = all(
    cell58_gates.values()
)


# ============================================================
# 27) Diagnostic output
# ============================================================

print(
    "=== EXERCISE-UNIT V2 TARGET DESIGN FEASIBILITY — REPAIRED AGAIN ==="
)


print(
    "\n=== EXISTING TARGET SYSTEM CAPABILITY ==="
)


print(
    "Active Targets:",
    len(
        target_evidence_records
    )
)


print(
    "Targets with >1 Part:",
    cell58_existing_multi_part_target_count
)


print(
    "Maximum active Target Part count:",
    cell58_active_target_max_part_count
)


print(
    "Active Part-count distribution:",
    dict(
        sorted(
            cell58_active_target_part_count_distribution.items()
        )
    )
)


print(
    "\n=== LEGACY EXERCISE PART BASELINE ==="
)


print(
    "Legacy Exercise Parts:",
    len(
        cell58_legacy_target_parts
    )
)


print(
    "Role counts:",
    dict(
        cell58_legacy_target_part_role_counts
    )
)


print(
    "Uniform role:",
    cell58_uniform_legacy_role
)


print(
    "Legacy full-document span failures:",
    len(
        cell58_legacy_full_document_span_failures
    )
)


if (
    cell58_legacy_full_document_span_failures
):

    for row in (
        cell58_legacy_full_document_span_failures[
            :10
        ]
    ):

        print(
            "-",
            row
        )


print(
    "\n=== TARGET PART BUILDER ==="
)


print(
    "Builder present:",
    cell58_target_part_builder_present
)


print(
    "Signature:",
    cell58_target_part_builder_signature
)


print(
    "Required args:",
    cell58_target_part_builder_required_args
)


print(
    "Optional args:",
    cell58_target_part_builder_optional_args
)


print(
    "Defaults:",
    cell58_target_part_builder_defaults
)


print(
    "Unknown required args:",
    cell58_unknown_required_builder_args
)


print(
    "Arguments groundable:",
    cell58_builder_arguments_groundable
)


print(
    "\n=== V2 EVIDENCE UNIVERSE ==="
)


print(
    "Document occurrences:",
    cell58_total_selected_document_occurrences
)


print(
    "Unique documents:",
    cell58_unique_selected_document_count
)


print(
    "Cross-case duplicate occurrences:",
    cell58_cross_case_duplicate_occurrences
)


print(
    "\n=== DOCUMENT PART PREFLIGHT ==="
)


print(
    "Parts built:",
    len(
        cell58_document_part_preflight
    )
)


print(
    "Build failures:",
    len(
        cell58_document_part_build_failures
    )
)


print(
    "Schema failures:",
    len(
        cell58_part_schema_failures
    )
)


print(
    "Integrity failures:",
    len(
        cell58_part_integrity_failures
    )
)


if (
    cell58_document_part_build_failures
):

    print(
        "\nFirst Part build failures:"
    )

    for row in (
        cell58_document_part_build_failures[
            :10
        ]
    ):

        print(
            "-",
            row
        )


if (
    cell58_part_schema_failures
):

    print(
        "\nFirst Part schema failures:"
    )

    for row in (
        cell58_part_schema_failures[
            :10
        ]
    ):

        print(
            "-",
            row
        )


if (
    cell58_part_integrity_failures
):

    print(
        "\nFirst Part integrity failures:"
    )

    for row in (
        cell58_part_integrity_failures[
            :10
        ]
    ):

        print(
            "-",
            row
        )


print(
    "\n=== FULL-UNIT CANDIDATE ==="
)


print(
    "Cases constructed:",
    len(
        cell58_full_unit_candidate_rows
    )
)


print(
    "Full-unit build failures:",
    len(
        cell58_full_unit_build_failures
    )
)


print(
    "Order/coverage failures:",
    len(
        cell58_full_unit_integrity_failures
    )
)


print(
    "Part-count distribution:",
    dict(
        sorted(
            cell58_candidate_part_count_distribution.items()
        )
    )
)


print(
    "Token summary:",
    cell58_token_summary
)


print(
    "\n=== CANDIDATE TARGET ID ==="
)


print(
    "Candidate Target IDs:",
    len(
        cell58_candidate_target_id_rows
    )
)


print(
    "Build failures:",
    len(
        cell58_candidate_target_id_failures
    )
)


print(
    "Independent recomputation failures:",
    len(
        cell58_candidate_target_id_recompute_failures
    )
)


print(
    "Duplicate IDs:",
    len(
        cell58_duplicate_candidate_target_ids
    )
)


print(
    "Collisions with active Targets:",
    len(
        cell58_candidate_target_id_collisions_with_active
    )
)


if (
    cell58_candidate_target_id_failures
):

    print(
        "\nFirst Candidate Target-ID failures:"
    )

    for row in (
        cell58_candidate_target_id_failures[
            :10
        ]
    ):

        print(
            "-",
            row
        )


print(
    "\n=== LARGEST CANDIDATES ==="
)


for row in (
    cell58_largest_candidate_rows[
        :10
    ]
):

    print(
        "- parts:",
        row[
            "candidate_part_count"
        ],
        "| tokens:",
        row[
            "candidate_total_tokens"
        ],
        "| chars:",
        row[
            "candidate_total_chars"
        ],
        "| case:",
        row[
            "golden_case_id"
        ],
        "| heading:",
        row[
            "heading_path"
        ],
    )


print(
    "\n=== STRATEGY ASSESSMENT ==="
)


print(
    "Technical feasibility:",
    cell58_full_unit_strategy_technical_feasibility
)


print(
    "Benchmark status:",
    cell58_full_unit_strategy_benchmark_status
)


print(
    "\n=== ACTIVE COUNTS ==="
)


print(
    "Golden:",
    len(
        golden_case_records
    )
)


print(
    "Tasks:",
    len(
        task_draft_plan_rows
    )
)


print(
    "Targets:",
    len(
        target_evidence_records
    )
)


print(
    "Pilots:",
    len(
        pilot_selection_rows
    )
)


print(
    "\n=== CODE CELL 58 GATES ==="
)


for gate_name, passed in (
    cell58_gates.items()
):

    print(
        gate_name,
        ":",
        "PASS"
        if passed
        else "FAIL"
    )


print(
    "\nFAILED GATES:",
    [
        gate_name

        for gate_name, passed
        in cell58_gates.items()

        if not passed
    ]
)


print(
    "\nCODE CELL 58 GATE:",
    "PASS"
    if cell_58_pass
    else "FAIL"
)


assert cell_58_pass, (
    "Exercise-unit V2 Target feasibility remains unresolved. "
    "Do not materialize V2 Targets. Use the explicit FAILED "
    "GATES and diagnostic sections from this SAME Cell 58 "
    "to repair the exact remaining contract issue."
)
```

<h3>Freeze the V2 Exercise Gold-Target Semantics</h3>

<p>
Cell 58 ثابت کرد که کل Exercise Unit از نظر فنی می‌تواند با همان Target Contract موجود
به‌صورت Ordered Multi-Part Evidence نمایش داده شود.
</p>

<p>
اما «قابل نمایش بودن» به‌تنهایی مشخص نمی‌کند که Gold Target چه معنایی باید داشته باشد.
در این Cell سه مفهوم از یکدیگر جدا می‌شوند:
</p>

<ul>
<li><strong>Gold Evidence Boundary</strong>: مرز دقیق محتوایی که Exercise یا Self-Assessment را تشکیل می‌دهد؛</li>
<li><strong>Retrieval Target Representation</strong>: نمایش آن مرز برای ارزیابی Retrieval؛</li>
<li><strong>Question-Specific Context</strong>: زیرمجموعه‌ای از Evidence که بعداً ممکن است برای پاسخ به یک سؤال مشخص کافی باشد.</li>
</ul>

<p>
قاعده‌ی پیشنهادی برای Audit این است:
</p>

<p>
<strong>
Gold Evidence Boundary = کل Exercise Unit adjudicated،
اما Question-Specific Context الزاماً برابر با کل Unit نیست.
</strong>
</p>

<p>
در نتیجه خارج شدن از Exercise Unit یک Boundary Error است،
ولی لازم نیست هر سؤال آینده همه‌ی Partهای داخل Unit را در Context نهایی مصرف کند.
این تصمیم باید بعداً در Question/Context Evaluation و بر اساس خود سؤال انجام شود.
</p>

<p>
این Cell همچنین بررسی می‌کند که Schema فعلی Target Evidence واقعاً برای Multi-Part Gold Target
سازگار است و آیا Role مربوط به Partهای V2 باید از Legacy
<code>EXPLICIT_BOOK_REGION</code>
جدا شود.
</p>

```python
# Code Cell 59

# ============================================================
# Exercise V2 Gold-Target Semantic Decision
#
# Goal:
#   freeze the SEMANTIC meaning of the V2 Exercise Target
#   before materializing any Target Evidence records.
#
# Core distinction:
#
#   GOLD EVIDENCE BOUNDARY
#       = complete adjudicated Exercise Unit
#
#   RETRIEVAL TARGET REPRESENTATION
#       = ordered real-document Parts covering that exact unit
#
#   QUESTION-SPECIFIC CONTEXT
#       = deferred; may later be a proper subset of the Gold
#         boundary depending on the actual drafted question
#
# READ-ONLY / CONTRACT FREEZE ONLY.
#
# NO V2 Target materialization.
# NO active mutation.
# NO Case activation.
# NO Task activation.
# NO Pilot rebuild.
# NO question generation.
# NO artifact writes.
# ============================================================


# ============================================================
# 1) Active/staging snapshots
# ============================================================

cell59_target_snapshot_before = hashlib.sha256(
    stable_json_key(
        target_evidence_records
    ).encode("utf-8")
).hexdigest()


cell59_task_snapshot_before = hashlib.sha256(
    stable_json_key(
        task_draft_plan_rows
    ).encode("utf-8")
).hexdigest()


cell59_golden_snapshot_before = hashlib.sha256(
    stable_json_key(
        golden_case_records
    ).encode("utf-8")
).hexdigest()


cell59_pilot_snapshot_before = hashlib.sha256(
    stable_json_key(
        pilot_selection_rows
    ).encode("utf-8")
).hexdigest()


cell59_case_staging_snapshot_before = hashlib.sha256(
    stable_json_key(
        staged_exercise_v2_golden_case_records
    ).encode("utf-8")
).hexdigest()


cell59_task_staging_snapshot_before = hashlib.sha256(
    stable_json_key(
        staged_exercise_v2_task_draft_plan_rows
    ).encode("utf-8")
).hexdigest()


# ============================================================
# 2) Inspect the actual Exercise Task-definition source
# ============================================================

cell59_exercise_task_source_snippets = []


for code_cell in (
    cell49_code_cells
):

    source_lines = (
        code_cell[
            "source"
        ].splitlines()
    )


    for line_index, line in enumerate(
        source_lines
    ):

        if (
            "EXERCISE_SELF_ASSESSMENT_RETRIEVAL"
            not in
            line
        ):

            continue


        start_index = max(
            0,
            line_index - 12,
        )


        end_index = min(
            len(
                source_lines
            ),
            line_index + 45,
        )


        cell59_exercise_task_source_snippets.append(
            {
                "cell_index":
                    code_cell[
                        "cell_index"
                    ],

                "line_index":
                    line_index,

                "snippet":
                    "\n".join(
                        source_lines[
                            start_index:
                            end_index
                        ]
                    ),
            }
        )


# ============================================================
# 3) Verify all 38 staged V2 Tasks retain the intended
# Exercise retrieval semantics
# ============================================================

cell59_staged_task_semantic_failures = []


CELL59_EXPECTED_TASK_SEMANTICS = {
    "source_type":
        "book",

    "stratum":
        "exercise_or_self_assessment",

    "task_family":
        "EXERCISE_SELF_ASSESSMENT_RETRIEVAL",

    "question_archetype":
        "STRUCTURE_AWARE_RETRIEVAL",

    "evidence_scope":
        "BOOK_EXERCISE_UNIT",

    "evidence_target_method":
        "USE_EXPLICIT_BOOK_EXERCISE_UNIT",

    "authority_expectation":
        "BOOK_PRIMARY_EVIDENCE",

    "context_expectation":
        "EXERCISE_BOUNDARY_PRESERVED",

    "required_dimensions":
        [
            "retrieval_hit",
            "rank",
            "evidence_completeness",
            "citation_locator_correctness",
            "unnecessary_context",
        ],
}


for task_row in (
    staged_exercise_v2_task_draft_plan_rows
):

    for field_name, expected_value in (
        CELL59_EXPECTED_TASK_SEMANTICS.items()
    ):

        if (
            task_row.get(
                field_name
            )
            !=
            expected_value
        ):

            cell59_staged_task_semantic_failures.append(
                {
                    "golden_case_id":
                        task_row[
                            "golden_case_id"
                        ],

                    "field":
                        field_name,

                    "actual":
                        task_row.get(
                            field_name
                        ),

                    "expected":
                        expected_value,
                }
            )


# ============================================================
# 4) Verify candidate Full-unit Targets equal the exact
# adjudicated Case evidence boundary
# ============================================================

cell59_candidate_by_case_id = {
    row[
        "golden_case_id"
    ]:
        row

    for row
    in cell58_full_unit_candidate_rows
}


cell59_case_boundary_failures = []


for staged_case in (
    staged_exercise_v2_golden_case_records
):

    case_id = (
        staged_case[
            "golden_case_id"
        ]
    )


    candidate = (
        cell59_candidate_by_case_id.get(
            case_id
        )
    )


    if (
        candidate
        is None
    ):

        cell59_case_boundary_failures.append(
            {
                "golden_case_id":
                    case_id,

                "reason":
                    "FULL_UNIT_CANDIDATE_MISSING",
            }
        )

        continue


    case_document_ids = list(
        staged_case[
            "evidence_document_ids"
        ]
    )


    candidate_document_ids = [
        part[
            "document_id"
        ]

        for part
        in candidate[
            "candidate_parts"
        ]
    ]


    if (
        candidate_document_ids
        !=
        case_document_ids
    ):

        cell59_case_boundary_failures.append(
            {
                "golden_case_id":
                    case_id,

                "reason":
                    "CANDIDATE_DOES_NOT_EQUAL_CASE_EVIDENCE_BOUNDARY",
            }
        )


# ============================================================
# 5) Validate candidate token totals against the frozen V2
# Case structural snapshot where available
# ============================================================

cell59_case_token_reconciliation_failures = []

cell59_case_token_reconciliation_available = 0


for staged_case in (
    staged_exercise_v2_golden_case_records
):

    case_id = (
        staged_case[
            "golden_case_id"
        ]
    )


    candidate = (
        cell59_candidate_by_case_id[
            case_id
        ]
    )


    candidate_snapshot = (
        staged_case.get(
            "candidate_snapshot",
            {}
        )
    )


    if (
        "exercise_core_token_count"
        not in
        candidate_snapshot
    ):

        continue


    cell59_case_token_reconciliation_available += 1


    frozen_token_count = (
        candidate_snapshot[
            "exercise_core_token_count"
        ]
    )


    candidate_token_count = (
        candidate[
            "candidate_total_tokens"
        ]
    )


    if (
        frozen_token_count
        !=
        candidate_token_count
    ):

        cell59_case_token_reconciliation_failures.append(
            {
                "golden_case_id":
                    case_id,

                "frozen_exercise_core_tokens":
                    frozen_token_count,

                "candidate_target_tokens":
                    candidate_token_count,
            }
        )


# ============================================================
# 6) Target schema compatibility audit
#
# Existing active Targets already include multi-part records.
# Candidate V2 targets use the same Part schema and the same
# target_evidence_id builder.
# ============================================================

cell59_active_target_schema_versions = Counter(
    row[
        "target_schema_version"
    ]

    for row
    in target_evidence_records
)


cell59_active_multi_part_schema_versions = Counter(
    row[
        "target_schema_version"
    ]

    for row
    in target_evidence_records

    if (
        len(
            row[
                "target_parts"
            ]
        )
        >
        1
    )
)


cell59_target_schema_compatible = (
    len(
        cell59_active_target_schema_versions
    )
    ==
    1

    and

    TARGET_EVIDENCE_SCHEMA_VERSION
    in
    cell59_active_target_schema_versions

    and

    cell58_existing_multi_part_target_count
    >
    0

    and

    len(
        cell58_part_schema_failures
    )
    ==
    0

    and

    len(
        cell58_candidate_target_id_failures
    )
    ==
    0

    and

    len(
        cell58_candidate_target_id_recompute_failures
    )
    ==
    0
)


# ============================================================
# 7) Audit active Part-role vocabulary and method->role
# relationships.
#
# We do not blindly reuse EXPLICIT_BOOK_REGION for a V2
# Exercise-unit Part role.
# ============================================================

cell59_active_part_role_counts = Counter(
    part.get(
        "role"
    )

    for target_row
    in target_evidence_records

    for part
    in target_row[
        "target_parts"
    ]
)


cell59_method_to_role_counts = {}


for target_row in (
    target_evidence_records
):

    method = (
        target_row[
            "evidence_target_method"
        ]
    )


    if (
        method
        not in
        cell59_method_to_role_counts
    ):

        cell59_method_to_role_counts[
            method
        ] = Counter()


    for part in (
        target_row[
            "target_parts"
        ]
    ):

        cell59_method_to_role_counts[
            method
        ][
            part.get(
                "role"
            )
        ] += 1


cell59_legacy_exercise_method_role_counts = (
    cell59_method_to_role_counts.get(
        "USE_EXPLICIT_BOOK_REGION",
        Counter(),
    )
)


# ============================================================
# 8) Freeze a scoped V2 Exercise Part role
#
# This is a semantic label only.
# It does NOT alter Part identity rules or fabricate evidence.
# ============================================================

EXERCISE_V2_TARGET_PART_ROLE = (
    "EXPLICIT_BOOK_EXERCISE_UNIT"
)


cell59_v2_role_is_distinct_from_legacy = (
    EXERCISE_V2_TARGET_PART_ROLE
    !=
    "EXPLICIT_BOOK_REGION"
)


# ============================================================
# 9) Define the Gold Target semantic decision
# ============================================================

EXERCISE_V2_GOLD_TARGET_BOUNDARY = (
    "FULL_ADJUDICATED_EXERCISE_UNIT"
)


EXERCISE_V2_GOLD_TARGET_REPRESENTATION = (
    "ORDERED_REAL_DOCUMENT_PARTS"
)


EXERCISE_V2_TARGET_SELECTION_BASIS = (
    "EXACT_ADJUDICATED_EXERCISE_UNIT"
)


EXERCISE_V2_QUESTION_CONTEXT_POLICY = (
    "QUESTION_SPECIFIC_SUBSET_MAY_BE_SELECTED_LATER_"
    "BUT_MUST_REMAIN_WITHIN_GOLD_EXERCISE_BOUNDARY"
)


# ============================================================
# 10) Explicitly separate four evaluation meanings
# ============================================================

EXERCISE_V2_EVALUATION_SEMANTICS = {
    "retrieval_hit": (
        "EVALUATE_RETRIEVAL_AGAINST_THE_ADJUDICATED_"
        "EXERCISE_UNIT_EVIDENCE_SET_NOT_AN_ARBITRARY_"
        "LEGACY_RECORD"
    ),

    "rank": (
        "RANK_EVALUATION_MUST_PRESERVE_MULTI_PART_GOLD_"
        "SEMANTICS_AND_MUST_NOT_COLLAPSE_THE_UNIT_TO_ONE_"
        "LEGACY_RECORD"
    ),

    "evidence_completeness": (
        "MEASURE_COVERAGE_RELATIVE_TO_THE_ADJUDICATED_"
        "EXERCISE_UNIT_WHEN_THE_TASK_REQUIRES_UNIT_LEVEL_"
        "COMPLETENESS"
    ),

    "unnecessary_context": (
        "EVIDENCE_OUTSIDE_THE_ADJUDICATED_EXERCISE_UNIT_"
        "IS_OUT_OF_BOUNDARY; EVIDENCE_INSIDE_THE_UNIT_IS_"
        "NOT_AUTOMATICALLY_REQUIRED_FOR_EVERY_FUTURE_QUESTION"
    ),

    "citation_locator_correctness": (
        "PRESERVE_AND_JUDGE_DOCUMENT_LEVEL_LOCATORS_FOR_"
        "THE_ACTUAL_EVIDENCE_PARTS_USED"
    ),
}


# ============================================================
# 11) Important distinction:
#
# Full Gold Boundary does NOT imply full generation context.
#
# Therefore no question-dependent Target subset is selected
# here.
# ============================================================

cell59_question_specific_context_selection_status = (
    "NOT_STARTED_REQUIRES_ACTUAL_QUESTION"
)


cell59_question_specific_context_parts = None


cell59_generation_context_not_equated_to_full_gold = (
    cell59_question_specific_context_parts
    is None

    and

    cell59_question_specific_context_selection_status
    ==
    "NOT_STARTED_REQUIRES_ACTUAL_QUESTION"
)


# ============================================================
# 12) Structural size is descriptive, not an exclusion rule
#
# Cell 58 showed:
#   max = 1295 tokens
#   max = 153 Parts
#
# No arbitrary size threshold is introduced here.
# ============================================================

cell59_full_unit_size_profile = {
    "part_count_distribution":
        dict(
            sorted(
                cell58_candidate_part_count_distribution.items()
            )
        ),

    "token_summary":
        cell52_safe_clone(
            cell58_token_summary
        ),

    "largest_part_count":
        max(
            row[
                "candidate_part_count"
            ]

            for row
            in cell58_full_unit_candidate_rows
        ),

    "largest_token_count":
        max(
            row[
                "candidate_total_tokens"
            ]

            for row
            in cell58_full_unit_candidate_rows
        ),

    "arbitrary_size_exclusion_threshold":
        None,
}


# ============================================================
# 13) Validate every V2 candidate is:
#
# - inside the Gold boundary
# - ordered
# - nonempty
# - real-document based
# - locator preserving
# - no synthetic aggregation
# ============================================================

cell59_gold_representation_failures = []


for candidate in (
    cell58_full_unit_candidate_rows
):

    reasons = []


    parts = (
        candidate[
            "candidate_parts"
        ]
    )


    if not (
        parts
    ):

        reasons.append(
            "EMPTY_TARGET_PARTS"
        )


    for part in (
        parts
    ):

        document_id = (
            part.get(
                "document_id"
            )
        )


        if (
            document_id
            not in
            corpus_target_index
        ):

            reasons.append(
                "NON_CORPUS_DOCUMENT_PART"
            )


        if (
            part.get(
                "locator"
            )
            is None
        ):

            reasons.append(
                "LOCATOR_MISSING"
            )


        if (
            part.get(
                "target_text_sha256"
            )
            is None
        ):

            reasons.append(
                "TARGET_TEXT_SHA_MISSING"
            )


    if (
        reasons
    ):

        cell59_gold_representation_failures.append(
            {
                "golden_case_id":
                    candidate[
                        "golden_case_id"
                    ],

                "reasons":
                    sorted(
                        set(
                            reasons
                        )
                    ),
            }
        )


# ============================================================
# 14) Freeze semantic contract
# ============================================================

EXERCISE_V2_GOLD_TARGET_SEMANTIC_CONTRACT_V1_DRAFT = {
    "version":
        "exercise_v2_gold_target_semantic_contract_v1_draft",

    "status":
        "SEMANTICS_FROZEN_TARGET_MATERIALIZATION_NOT_STARTED",

    "target_schema_version":
        TARGET_EVIDENCE_SCHEMA_VERSION,

    "gold_evidence_boundary":
        EXERCISE_V2_GOLD_TARGET_BOUNDARY,

    "target_representation":
        EXERCISE_V2_GOLD_TARGET_REPRESENTATION,

    "target_selection_basis":
        EXERCISE_V2_TARGET_SELECTION_BASIS,

    "evidence_scope":
        EXERCISE_V2_TASK_EVIDENCE_SCOPE,

    "evidence_target_method":
        EXERCISE_V2_TASK_EVIDENCE_TARGET_METHOD,

    "part_role":
        EXERCISE_V2_TARGET_PART_ROLE,

    "boundary_semantics": {
        "inside_unit":
            "ELIGIBLE_GOLD_EVIDENCE",

        "outside_unit":
            "OUTSIDE_GOLD_BOUNDARY",
    },

    "question_context_policy":
        EXERCISE_V2_QUESTION_CONTEXT_POLICY,

    "evaluation_semantics":
        cell52_safe_clone(
            EXERCISE_V2_EVALUATION_SEMANTICS
        ),

    "full_unit_size_profile":
        cell52_safe_clone(
            cell59_full_unit_size_profile
        ),

    "synthetic_aggregation":
        False,

    "semantic_subselection_at_target_build_time":
        False,

    "question_specific_context_selection":
        "DEFERRED_UNTIL_ACTUAL_QUESTION",

    "target_materialization_status":
        "NOT_STARTED",

    "task_ready_transition_status":
        "NOT_STARTED",

    "case_activation_status":
        "NOT_STARTED",

    "task_activation_status":
        "NOT_STARTED",

    "pilot_rebuild_status":
        "NOT_STARTED",

    "question_generation_status":
        "NOT_STARTED",

    "artifact_write_status":
        "NONE",
}


# ============================================================
# 15) No-mutation recheck
# ============================================================

cell59_target_snapshot_after = hashlib.sha256(
    stable_json_key(
        target_evidence_records
    ).encode("utf-8")
).hexdigest()


cell59_task_snapshot_after = hashlib.sha256(
    stable_json_key(
        task_draft_plan_rows
    ).encode("utf-8")
).hexdigest()


cell59_golden_snapshot_after = hashlib.sha256(
    stable_json_key(
        golden_case_records
    ).encode("utf-8")
).hexdigest()


cell59_pilot_snapshot_after = hashlib.sha256(
    stable_json_key(
        pilot_selection_rows
    ).encode("utf-8")
).hexdigest()


cell59_case_staging_snapshot_after = hashlib.sha256(
    stable_json_key(
        staged_exercise_v2_golden_case_records
    ).encode("utf-8")
).hexdigest()


cell59_task_staging_snapshot_after = hashlib.sha256(
    stable_json_key(
        staged_exercise_v2_task_draft_plan_rows
    ).encode("utf-8")
).hexdigest()


# ============================================================
# 16) Validation gates
# ============================================================

cell59_gates = {
    "exact_38_staged_v2_tasks_preserve_exercise_retrieval_semantics":
        (
            len(
                staged_exercise_v2_task_draft_plan_rows
            )
            ==
            38

            and

            len(
                cell59_staged_task_semantic_failures
            )
            ==
            0
        ),

    "actual_exercise_task_definition_source_was_inspected":
        (
            len(
                cell59_exercise_task_source_snippets
            )
            >
            0
        ),

    "exact_38_full_unit_candidates_equal_adjudicated_case_boundaries":
        (
            len(
                cell58_full_unit_candidate_rows
            )
            ==
            38

            and

            len(
                cell59_case_boundary_failures
            )
            ==
            0
        ),

    "frozen_exercise_core_token_counts_reconcile_where_available":
        (
            len(
                cell59_case_token_reconciliation_failures
            )
            ==
            0
        ),

    "current_target_schema_is_multi_part_compatible":
        cell59_target_schema_compatible,

    "legacy_region_role_is_not_silently_reused_as_v2_unit_role":
        cell59_v2_role_is_distinct_from_legacy,

    "gold_evidence_boundary_is_full_adjudicated_exercise_unit":
        (
            EXERCISE_V2_GOLD_TARGET_BOUNDARY
            ==
            "FULL_ADJUDICATED_EXERCISE_UNIT"
        ),

    "gold_target_representation_is_ordered_real_document_parts":
        (
            EXERCISE_V2_GOLD_TARGET_REPRESENTATION
            ==
            "ORDERED_REAL_DOCUMENT_PARTS"
        ),

    "target_selection_basis_is_exact_adjudicated_exercise_unit":
        (
            EXERCISE_V2_TARGET_SELECTION_BASIS
            ==
            "EXACT_ADJUDICATED_EXERCISE_UNIT"
        ),

    "question_specific_context_is_not_falsely_equated_to_full_gold_boundary":
        cell59_generation_context_not_equated_to_full_gold,

    "all_38_gold_candidates_use_real_locator_preserving_corpus_parts":
        (
            len(
                cell59_gold_representation_failures
            )
            ==
            0
        ),

    "no_arbitrary_size_exclusion_threshold_introduced":
        (
            cell59_full_unit_size_profile[
                "arbitrary_size_exclusion_threshold"
            ]
            is None
        ),

    "target_materialization_not_started":
        (
            EXERCISE_V2_GOLD_TARGET_SEMANTIC_CONTRACT_V1_DRAFT[
                "target_materialization_status"
            ]
            ==
            "NOT_STARTED"
        ),

    "task_ready_transition_not_started":
        (
            EXERCISE_V2_GOLD_TARGET_SEMANTIC_CONTRACT_V1_DRAFT[
                "task_ready_transition_status"
            ]
            ==
            "NOT_STARTED"
        ),

    "case_activation_not_started":
        (
            EXERCISE_V2_GOLD_TARGET_SEMANTIC_CONTRACT_V1_DRAFT[
                "case_activation_status"
            ]
            ==
            "NOT_STARTED"
        ),

    "task_activation_not_started":
        (
            EXERCISE_V2_GOLD_TARGET_SEMANTIC_CONTRACT_V1_DRAFT[
                "task_activation_status"
            ]
            ==
            "NOT_STARTED"
        ),

    "pilot_rebuild_not_started":
        (
            EXERCISE_V2_GOLD_TARGET_SEMANTIC_CONTRACT_V1_DRAFT[
                "pilot_rebuild_status"
            ]
            ==
            "NOT_STARTED"
        ),

    "question_generation_not_started":
        (
            EXERCISE_V2_GOLD_TARGET_SEMANTIC_CONTRACT_V1_DRAFT[
                "question_generation_status"
            ]
            ==
            "NOT_STARTED"
        ),

    "all_active_and_existing_staging_registries_content_unchanged":
        (
            cell59_target_snapshot_before
            ==
            cell59_target_snapshot_after

            and

            cell59_task_snapshot_before
            ==
            cell59_task_snapshot_after

            and

            cell59_golden_snapshot_before
            ==
            cell59_golden_snapshot_after

            and

            cell59_pilot_snapshot_before
            ==
            cell59_pilot_snapshot_after

            and

            cell59_case_staging_snapshot_before
            ==
            cell59_case_staging_snapshot_after

            and

            cell59_task_staging_snapshot_before
            ==
            cell59_task_staging_snapshot_after

            and

            len(
                golden_case_records
            )
            ==
            724

            and

            len(
                task_draft_plan_rows
            )
            ==
            724

            and

            len(
                target_evidence_records
            )
            ==
            572

            and

            len(
                pilot_selection_rows
            )
            ==
            23
        ),

    "no_artifact_write":
        (
            EXERCISE_V2_GOLD_TARGET_SEMANTIC_CONTRACT_V1_DRAFT[
                "artifact_write_status"
            ]
            ==
            "NONE"
        ),
}


cell_59_pass = all(
    cell59_gates.values()
)


# ============================================================
# 17) Output
# ============================================================

print(
    "=== EXERCISE V2 GOLD-TARGET SEMANTIC DECISION ==="
)


print(
    "\n=== TASK SEMANTICS ==="
)


print(
    "Staged V2 Tasks:",
    len(
        staged_exercise_v2_task_draft_plan_rows
    )
)


print(
    "Task semantic failures:",
    len(
        cell59_staged_task_semantic_failures
    )
)


print(
    "Exercise Task source snippets found:",
    len(
        cell59_exercise_task_source_snippets
    )
)


print(
    "\n=== GOLD BOUNDARY RECONCILIATION ==="
)


print(
    "Full-unit candidates:",
    len(
        cell58_full_unit_candidate_rows
    )
)


print(
    "Boundary failures:",
    len(
        cell59_case_boundary_failures
    )
)


print(
    "Token reconciliation available:",
    cell59_case_token_reconciliation_available,
    "/ 38",
)


print(
    "Token reconciliation failures:",
    len(
        cell59_case_token_reconciliation_failures
    )
)


if (
    cell59_case_token_reconciliation_failures
):

    print(
        "\nFirst token reconciliation failures:"
    )

    for row in (
        cell59_case_token_reconciliation_failures[
            :10
        ]
    ):

        print(
            "-",
            row,
        )


print(
    "\n=== TARGET SCHEMA COMPATIBILITY ==="
)


print(
    "Active Target schema versions:",
    dict(
        cell59_active_target_schema_versions
    )
)


print(
    "Multi-part Target schema versions:",
    dict(
        cell59_active_multi_part_schema_versions
    )
)


print(
    "Schema compatible:",
    cell59_target_schema_compatible
)


print(
    "\n=== PART ROLE AUDIT ==="
)


print(
    "Legacy Exercise method role counts:",
    dict(
        cell59_legacy_exercise_method_role_counts
    )
)


print(
    "V2 scoped Part role:",
    EXERCISE_V2_TARGET_PART_ROLE
)


print(
    "\n=== FROZEN GOLD-TARGET SEMANTICS ==="
)


print(
    "Gold Evidence Boundary:",
    EXERCISE_V2_GOLD_TARGET_BOUNDARY
)


print(
    "Target Representation:",
    EXERCISE_V2_GOLD_TARGET_REPRESENTATION
)


print(
    "Selection Basis:",
    EXERCISE_V2_TARGET_SELECTION_BASIS
)


print(
    "Evidence Scope:",
    EXERCISE_V2_TASK_EVIDENCE_SCOPE
)


print(
    "Evidence Method:",
    EXERCISE_V2_TASK_EVIDENCE_TARGET_METHOD
)


print(
    "Question Context Policy:",
    EXERCISE_V2_QUESTION_CONTEXT_POLICY
)


print(
    "\n=== EVALUATION SEMANTICS ==="
)


for dimension, meaning in (
    EXERCISE_V2_EVALUATION_SEMANTICS.items()
):

    print(
        "-",
        dimension,
        "=>",
        meaning,
    )


print(
    "\n=== FULL-UNIT SIZE PROFILE ==="
)


print(
    "Part-count distribution:",
    cell59_full_unit_size_profile[
        "part_count_distribution"
    ]
)


print(
    "Token summary:",
    cell59_full_unit_size_profile[
        "token_summary"
    ]
)


print(
    "Largest Part count:",
    cell59_full_unit_size_profile[
        "largest_part_count"
    ]
)


print(
    "Largest Token count:",
    cell59_full_unit_size_profile[
        "largest_token_count"
    ]
)


print(
    "Arbitrary exclusion threshold:",
    cell59_full_unit_size_profile[
        "arbitrary_size_exclusion_threshold"
    ]
)


print(
    "\nGold representation failures:",
    len(
        cell59_gold_representation_failures
    )
)


print(
    "\n=== ACTIVE COUNTS ==="
)


print(
    "Golden:",
    len(
        golden_case_records
    )
)


print(
    "Tasks:",
    len(
        task_draft_plan_rows
    )
)


print(
    "Targets:",
    len(
        target_evidence_records
    )
)


print(
    "Pilots:",
    len(
        pilot_selection_rows
    )
)


print(
    "\n=== CODE CELL 59 GATES ==="
)


for gate_name, passed in (
    cell59_gates.items()
):

    print(
        gate_name,
        ":",
        "PASS"
        if passed
        else "FAIL"
    )


print(
    "\nFAILED GATES:",
    [
        gate_name

        for gate_name, passed
        in cell59_gates.items()

        if not passed
    ]
)


print(
    "\nCODE CELL 59 GATE:",
    "PASS"
    if cell_59_pass
    else "FAIL"
)


assert cell_59_pass, (
    "Exercise V2 Gold-Target semantics are not yet safely "
    "resolved. Do not materialize V2 Targets. Repair this SAME "
    "Code Cell 59 until the Gold Exercise boundary, multi-part "
    "representation, schema compatibility and distinction from "
    "question-specific context are all explicit and validated."
)
```

```python
# Code Cell 60

# ============================================================
# V2 Target Record Assembly Substructure Contract Discovery
#
# Goal:
#   inspect the exact persisted + source-level construction
#   contract for all nested Target Evidence fields before
#   staging V2 multi-part Exercise Targets.
#
# READ-ONLY.
#
# NO V2 Target records.
# NO active mutation.
# NO Case activation.
# NO Task activation.
# NO Pilot rebuild.
# NO question generation.
# NO artifact writes.
# ============================================================


# ============================================================
# 1) Snapshot active / existing staging state
# ============================================================

cell60_target_snapshot_before = hashlib.sha256(
    stable_json_key(
        target_evidence_records
    ).encode("utf-8")
).hexdigest()


cell60_task_snapshot_before = hashlib.sha256(
    stable_json_key(
        task_draft_plan_rows
    ).encode("utf-8")
).hexdigest()


cell60_golden_snapshot_before = hashlib.sha256(
    stable_json_key(
        golden_case_records
    ).encode("utf-8")
).hexdigest()


cell60_pilot_snapshot_before = hashlib.sha256(
    stable_json_key(
        pilot_selection_rows
    ).encode("utf-8")
).hexdigest()


cell60_case_staging_snapshot_before = hashlib.sha256(
    stable_json_key(
        staged_exercise_v2_golden_case_records
    ).encode("utf-8")
).hexdigest()


cell60_task_staging_snapshot_before = hashlib.sha256(
    stable_json_key(
        staged_exercise_v2_task_draft_plan_rows
    ).encode("utf-8")
).hexdigest()


# ============================================================
# 2) Target fields requiring exact assembly resolution
# ============================================================

CELL60_ASSEMBLY_FIELDS = [
    "target_text_spans",
    "target_locator_refs",
    "target_attribution_state",
    "target_authority_state",
    "target_selection_basis",
    "citation_target_status",
]


# ============================================================
# 3) Recursive structural profiler
#
# Does not expose or alter data.
# ============================================================

def cell60_structure_signature(
    value,
):

    if isinstance(
        value,
        dict,
    ):

        return {
            "type":
                "dict",

            "keys":
                sorted(
                    value.keys()
                ),

            "children":
                {
                    key:
                        cell60_structure_signature(
                            value[
                                key
                            ]
                        )

                    for key
                    in sorted(
                        value.keys()
                    )
                },
        }


    if isinstance(
        value,
        list,
    ):

        item_signatures = [
            stable_json_key(
                cell60_structure_signature(
                    item
                )
            )

            for item
            in value
        ]


        return {
            "type":
                "list",

            "length":
                len(
                    value
                ),

            "unique_item_structure_count":
                len(
                    set(
                        item_signatures
                    )
                ),

            "item_structures":
                sorted(
                    set(
                        item_signatures
                    )
                ),
        }


    return {
        "type":
            type(
                value
            ).__name__,
    }


# ============================================================
# 4) Legacy Exercise assembly field profiles
# ============================================================

cell60_legacy_field_profiles = {}


for field_name in (
    CELL60_ASSEMBLY_FIELDS
):

    values = [
        row[
            field_name
        ]

        for row
        in cell57_legacy_target_rows
    ]


    structure_signatures = [
        stable_json_key(
            cell60_structure_signature(
                value
            )
        )

        for value
        in values
    ]


    serialized_values = [
        stable_json_key(
            value
        )

        for value
        in values
    ]


    cell60_legacy_field_profiles[
        field_name
    ] = {
        "row_count":
            len(
                values
            ),

        "value_types":
            dict(
                Counter(
                    type(
                        value
                    ).__name__

                    for value
                    in values
                )
            ),

        "unique_structure_count":
            len(
                set(
                    structure_signatures
                )
            ),

        "unique_value_count":
            len(
                set(
                    serialized_values
                )
            ),

        "constant":
            (
                len(
                    set(
                        serialized_values
                    )
                )
                ==
                1
            ),

        "example_structure":
            (
                cell60_structure_signature(
                    values[
                        0
                    ]
                )
                if values
                else None
            ),
    }


# ============================================================
# 5) Profile same fields globally across 572 active Targets
#
# This reveals how multi-part Targets already represent these
# substructures.
# ============================================================

cell60_global_field_profiles = {}


for field_name in (
    CELL60_ASSEMBLY_FIELDS
):

    values = [
        row[
            field_name
        ]

        for row
        in target_evidence_records
    ]


    structure_signatures = [
        stable_json_key(
            cell60_structure_signature(
                value
            )
        )

        for value
        in values
    ]


    cell60_global_field_profiles[
        field_name
    ] = {
        "row_count":
            len(
                values
            ),

        "value_types":
            dict(
                Counter(
                    type(
                        value
                    ).__name__

                    for value
                    in values
                )
            ),

        "unique_structure_count":
            len(
                set(
                    structure_signatures
                )
            ),
    }


# ============================================================
# 6) Multi-part active Target subset
# ============================================================

cell60_active_multi_part_targets = [
    row

    for row
    in target_evidence_records

    if (
        len(
            row[
                "target_parts"
            ]
        )
        >
        1
    )
]


cell60_multi_part_field_profiles = {}


for field_name in (
    CELL60_ASSEMBLY_FIELDS
):

    values = [
        row[
            field_name
        ]

        for row
        in cell60_active_multi_part_targets
    ]


    structure_signatures = [
        stable_json_key(
            cell60_structure_signature(
                value
            )
        )

        for value
        in values
    ]


    cell60_multi_part_field_profiles[
        field_name
    ] = {
        "row_count":
            len(
                values
            ),

        "value_types":
            dict(
                Counter(
                    type(
                        value
                    ).__name__

                    for value
                    in values
                )
            ),

        "unique_structure_count":
            len(
                set(
                    structure_signatures
                )
            ),

        "example_structure":
            (
                cell60_structure_signature(
                    values[
                        0
                    ]
                )
                if values
                else None
            ),
    }


# ============================================================
# 7) Exact Text-Span projection discovery
#
# Determine whether each persisted target_text_spans item is
# an exact projection of its corresponding target_part.
# ============================================================

cell60_text_span_projection_failures = []

cell60_text_span_keys = None


for target_row in (
    target_evidence_records
):

    parts = (
        target_row[
            "target_parts"
        ]
    )


    spans = (
        target_row[
            "target_text_spans"
        ]
    )


    if (
        len(
            parts
        )
        !=
        len(
            spans
        )
    ):

        cell60_text_span_projection_failures.append(
            {
                "target_evidence_id":
                    target_row[
                        "target_evidence_id"
                    ],

                "reason":
                    "PART_SPAN_LENGTH_MISMATCH",
            }
        )

        continue


    for part, span in zip(
        parts,
        spans,
    ):

        if not isinstance(
            span,
            dict,
        ):

            cell60_text_span_projection_failures.append(
                {
                    "target_evidence_id":
                        target_row[
                            "target_evidence_id"
                        ],

                    "reason":
                        "SPAN_NOT_DICT",
                }
            )

            continue


        span_keys = set(
            span.keys()
        )


        if (
            cell60_text_span_keys
            is None
        ):

            cell60_text_span_keys = (
                span_keys
            )


        elif (
            span_keys
            !=
            cell60_text_span_keys
        ):

            cell60_text_span_projection_failures.append(
                {
                    "target_evidence_id":
                        target_row[
                            "target_evidence_id"
                        ],

                    "reason":
                        "SPAN_SCHEMA_NOT_UNIFORM",
                }
            )


        for key, value in (
            span.items()
        ):

            if (
                key
                not in
                part
            ):

                cell60_text_span_projection_failures.append(
                    {
                        "target_evidence_id":
                            target_row[
                                "target_evidence_id"
                            ],

                        "reason":
                            "SPAN_FIELD_NOT_PRESENT_IN_PART",

                        "field":
                            key,
                    }
                )

                continue


            if (
                stable_json_key(
                    value
                )
                !=
                stable_json_key(
                    part[
                        key
                    ]
                )
            ):

                cell60_text_span_projection_failures.append(
                    {
                        "target_evidence_id":
                            target_row[
                                "target_evidence_id"
                            ],

                        "reason":
                            "SPAN_VALUE_NOT_EXACT_PART_PROJECTION",

                        "field":
                            key,
                    }
                )


cell60_text_span_projection_resolved = (
    cell60_text_span_keys
    is not None

    and

    len(
        cell60_text_span_projection_failures
    )
    ==
    0
)


# ============================================================
# 8) Exact Locator-Ref projection discovery
# ============================================================

cell60_locator_projection_failures = []

cell60_locator_ref_keys = None


for target_row in (
    target_evidence_records
):

    parts = (
        target_row[
            "target_parts"
        ]
    )


    locator_refs = (
        target_row[
            "target_locator_refs"
        ]
    )


    if (
        len(
            parts
        )
        !=
        len(
            locator_refs
        )
    ):

        cell60_locator_projection_failures.append(
            {
                "target_evidence_id":
                    target_row[
                        "target_evidence_id"
                    ],

                "reason":
                    "PART_LOCATOR_LENGTH_MISMATCH",
            }
        )

        continue


    for part, locator_ref in zip(
        parts,
        locator_refs,
    ):

        if not isinstance(
            locator_ref,
            dict,
        ):

            cell60_locator_projection_failures.append(
                {
                    "target_evidence_id":
                        target_row[
                            "target_evidence_id"
                        ],

                    "reason":
                        "LOCATOR_REF_NOT_DICT",
                }
            )

            continue


        locator_keys = set(
            locator_ref.keys()
        )


        if (
            cell60_locator_ref_keys
            is None
        ):

            cell60_locator_ref_keys = (
                locator_keys
            )


        elif (
            locator_keys
            !=
            cell60_locator_ref_keys
        ):

            cell60_locator_projection_failures.append(
                {
                    "target_evidence_id":
                        target_row[
                            "target_evidence_id"
                        ],

                    "reason":
                        "LOCATOR_REF_SCHEMA_NOT_UNIFORM",
                }
            )


        for key, value in (
            locator_ref.items()
        ):

            if (
                key
                not in
                part
            ):

                cell60_locator_projection_failures.append(
                    {
                        "target_evidence_id":
                            target_row[
                                "target_evidence_id"
                            ],

                        "reason":
                            "LOCATOR_FIELD_NOT_PRESENT_IN_PART",

                        "field":
                            key,
                    }
                )

                continue


            if (
                stable_json_key(
                    value
                )
                !=
                stable_json_key(
                    part[
                        key
                    ]
                )
            ):

                cell60_locator_projection_failures.append(
                    {
                        "target_evidence_id":
                            target_row[
                                "target_evidence_id"
                            ],

                        "reason":
                            "LOCATOR_VALUE_NOT_EXACT_PART_PROJECTION",

                        "field":
                            key,
                    }
                )


cell60_locator_projection_resolved = (
    cell60_locator_ref_keys
    is not None

    and

    len(
        cell60_locator_projection_failures
    )
    ==
    0
)


# ============================================================
# 9) Authority-state derivation audit
#
# Legacy Exercise authority state was constant:
# {
#   task_authority_expectation: BOOK_PRIMARY_EVIDENCE,
#   authority_inference_performed: False
# }
#
# Validate this relationship globally where possible.
# ============================================================

cell60_authority_state_failures = []


for target_row in (
    target_evidence_records
):

    task_id = (
        target_row[
            "task_draft_id"
        ]
    )


    if (
        task_id
        not in
        task_draft_plans
    ):

        cell60_authority_state_failures.append(
            {
                "target_evidence_id":
                    target_row[
                        "target_evidence_id"
                    ],

                "reason":
                    "TASK_NOT_FOUND",
            }
        )

        continue


    task_row = (
        task_draft_plans[
            task_id
        ]
    )


    expected_state = {
        "task_authority_expectation":
            task_row[
                "authority_expectation"
            ],

        "authority_inference_performed":
            False,
    }


    if (
        stable_json_key(
            target_row[
                "target_authority_state"
            ]
        )
        !=
        stable_json_key(
            expected_state
        )
    ):

        cell60_authority_state_failures.append(
            {
                "target_evidence_id":
                    target_row[
                        "target_evidence_id"
                    ],

                "reason":
                    "AUTHORITY_STATE_FORMULA_MISMATCH",
            }
        )


cell60_authority_state_formula_resolved = (
    len(
        cell60_authority_state_failures
    )
    ==
    0
)


# ============================================================
# 10) Attribution-state structural relationship audit
#
# We intentionally do NOT assume its formula yet.
# First inspect:
# - type
# - key schema
# - relation to part count
# - relationship with each Part's attribution / speaker_state
# ============================================================

cell60_attribution_state_type_counts = Counter(
    type(
        row[
            "target_attribution_state"
        ]
    ).__name__

    for row
    in target_evidence_records
)


cell60_attribution_state_structure_counts = Counter(
    stable_json_key(
        cell60_structure_signature(
            row[
                "target_attribution_state"
            ]
        )
    )

    for row
    in target_evidence_records
)


cell60_legacy_attribution_examples = [
    {
        "target_evidence_id":
            row[
                "target_evidence_id"
            ],

        "target_attribution_state":
            cell52_safe_clone(
                row[
                    "target_attribution_state"
                ]
            ),

        "part_attributions":
            [
                cell52_safe_clone(
                    part[
                        "attribution"
                    ]
                )

                for part
                in row[
                    "target_parts"
                ]
            ],

        "part_speaker_states":
            [
                cell52_safe_clone(
                    part[
                        "speaker_state"
                    ]
                )

                for part
                in row[
                    "target_parts"
                ]
            ],
    }

    for row
    in cell57_legacy_target_rows[
        :5
    ]
]


cell60_multi_part_attribution_examples = [
    {
        "target_evidence_id":
            row[
                "target_evidence_id"
            ],

        "part_count":
            len(
                row[
                    "target_parts"
                ]
            ),

        "target_attribution_state":
            cell52_safe_clone(
                row[
                    "target_attribution_state"
                ]
            ),

        "part_attributions":
            [
                cell52_safe_clone(
                    part[
                        "attribution"
                    ]
                )

                for part
                in row[
                    "target_parts"
                ]
            ],

        "part_speaker_states":
            [
                cell52_safe_clone(
                    part[
                        "speaker_state"
                    ]
                )

                for part
                in row[
                    "target_parts"
                ]
            ],
    }

    for row
    in cell60_active_multi_part_targets[
        :5
    ]
]


# ============================================================
# 11) Search actual Notebook source for complete Target return
# assembly and nested subfield formulas.
# ============================================================

CELL60_SOURCE_ANCHORS = [
    '"target_text_spans"',
    '"target_locator_refs"',
    '"target_attribution_state"',
    '"target_authority_state"',
    '"target_selection_basis"',
    '"citation_target_status"',
]


cell60_assembly_source_snippets = []


for code_cell in (
    cell49_code_cells
):

    source_lines = (
        code_cell[
            "source"
        ].splitlines()
    )


    for line_index, line in enumerate(
        source_lines
    ):

        matched_anchors = [
            anchor

            for anchor
            in CELL60_SOURCE_ANCHORS

            if (
                anchor
                in
                line
            )
        ]


        if not (
            matched_anchors
        ):

            continue


        start_index = max(
            0,
            line_index - 20,
        )


        end_index = min(
            len(
                source_lines
            ),
            line_index + 55,
        )


        cell60_assembly_source_snippets.append(
            {
                "cell_index":
                    code_cell[
                        "cell_index"
                    ],

                "line_index":
                    line_index,

                "matched_anchors":
                    matched_anchors,

                "snippet":
                    "\n".join(
                        source_lines[
                            start_index:
                            end_index
                        ]
                    ),
            }
        )


# ============================================================
# 12) Discover relevant assembly helper callables
# ============================================================

cell60_candidate_helper_callables = []


for variable_name, value in list(
    globals().items()
):

    if not callable(
        value
    ):

        continue


    lowered_name = (
        variable_name.lower()
    )


    if not (
        "target"
        in
        lowered_name

        and

        (
            "attribution"
            in
            lowered_name

            or

            "authority"
            in
            lowered_name

            or

            "locator"
            in
            lowered_name

            or

            "span"
            in
            lowered_name
        )
    ):

        continue


    code_object = getattr(
        value,
        "__code__",
        None,
    )


    if (
        code_object
        is None
    ):

        continue


    cell60_candidate_helper_callables.append(
        {
            "name":
                variable_name,

            "args":
                list(
                    code_object.co_varnames[
                        :code_object.co_argcount
                    ]
                ),

            "referenced_names":
                list(
                    code_object.co_names
                ),
        }
    )


# ============================================================
# 13) Selection-basis and citation-status baseline
# ============================================================

cell60_legacy_selection_basis_counts = Counter(
    row[
        "target_selection_basis"
    ]

    for row
    in cell57_legacy_target_rows
)


cell60_legacy_citation_status_counts = Counter(
    row[
        "citation_target_status"
    ]

    for row
    in cell57_legacy_target_rows
)


cell60_global_selection_basis_counts = Counter(
    row[
        "target_selection_basis"
    ]

    for row
    in target_evidence_records
)


cell60_global_citation_status_counts = Counter(
    row[
        "citation_target_status"
    ]

    for row
    in target_evidence_records
)


# ============================================================
# 14) Assembly-contract resolution status
#
# At this stage:
# - text-span formula can be fully resolved by projection;
# - locator formula can be fully resolved by projection;
# - authority-state formula can be independently reproduced;
# - attribution formula is intentionally required to be
#   inspected from multi-part evidence/source before V2 build.
# ============================================================

cell60_resolved_subcontracts = {
    "target_text_spans":
        (
            "EXACT_ORDERED_PROJECTION_FROM_TARGET_PARTS"
            if cell60_text_span_projection_resolved
            else
            "UNRESOLVED"
        ),

    "target_locator_refs":
        (
            "EXACT_ORDERED_PROJECTION_FROM_TARGET_PARTS"
            if cell60_locator_projection_resolved
            else
            "UNRESOLVED"
        ),

    "target_authority_state":
        (
            "TASK_AUTHORITY_EXPECTATION_PLUS_NO_INFERENCE"
            if cell60_authority_state_formula_resolved
            else
            "UNRESOLVED"
        ),

    "target_attribution_state":
        "INSPECTED_NOT_YET_REBUILT",

    "target_selection_basis":
        "V2_VALUE_ALREADY_FROZEN_IN_CELL59",

    "citation_target_status":
        "REQUIRES_POST_BUILD_LOCATOR_CAPTURE_STATE",
}


EXERCISE_V2_TARGET_ASSEMBLY_DISCOVERY_V1_DRAFT = {
    "version":
        "exercise_v2_target_assembly_discovery_v1_draft",

    "status":
        "SUBSTRUCTURE_CONTRACT_INSPECTED_TARGET_BUILD_NOT_STARTED",

    "resolved_subcontracts":
        cell52_safe_clone(
            cell60_resolved_subcontracts
        ),

    "text_span_projection_keys":
        (
            sorted(
                cell60_text_span_keys
            )
            if cell60_text_span_keys
            else []
        ),

    "locator_projection_keys":
        (
            sorted(
                cell60_locator_ref_keys
            )
            if cell60_locator_ref_keys
            else []
        ),

    "authority_state_formula_resolved":
        cell60_authority_state_formula_resolved,

    "attribution_state_status":
        "SOURCE_AND_MULTI_PART_STRUCTURE_INSPECTED",

    "v2_target_materialization_status":
        "NOT_STARTED",

    "case_activation_status":
        "NOT_STARTED",

    "task_activation_status":
        "NOT_STARTED",

    "pilot_rebuild_status":
        "NOT_STARTED",

    "question_generation_status":
        "NOT_STARTED",

    "artifact_write_status":
        "NONE",
}


# ============================================================
# 15) No-mutation recheck
# ============================================================

cell60_target_snapshot_after = hashlib.sha256(
    stable_json_key(
        target_evidence_records
    ).encode("utf-8")
).hexdigest()


cell60_task_snapshot_after = hashlib.sha256(
    stable_json_key(
        task_draft_plan_rows
    ).encode("utf-8")
).hexdigest()


cell60_golden_snapshot_after = hashlib.sha256(
    stable_json_key(
        golden_case_records
    ).encode("utf-8")
).hexdigest()


cell60_pilot_snapshot_after = hashlib.sha256(
    stable_json_key(
        pilot_selection_rows
    ).encode("utf-8")
).hexdigest()


cell60_case_staging_snapshot_after = hashlib.sha256(
    stable_json_key(
        staged_exercise_v2_golden_case_records
    ).encode("utf-8")
).hexdigest()


cell60_task_staging_snapshot_after = hashlib.sha256(
    stable_json_key(
        staged_exercise_v2_task_draft_plan_rows
    ).encode("utf-8")
).hexdigest()


# ============================================================
# 16) Gates
# ============================================================

cell60_gates = {
    "exact_38_legacy_exercise_targets_inspected":
        (
            len(
                cell57_legacy_target_rows
            )
            ==
            38
        ),

    "exact_270_existing_multi_part_targets_available_for_contract_comparison":
        (
            len(
                cell60_active_multi_part_targets
            )
            ==
            270
        ),

    "target_text_spans_are_exact_ordered_part_projection":
        cell60_text_span_projection_resolved,

    "target_locator_refs_are_exact_ordered_part_projection":
        cell60_locator_projection_resolved,

    "target_authority_state_formula_reproduces_active_targets":
        cell60_authority_state_formula_resolved,

    "attribution_state_structure_profiled_for_active_and_multi_part_targets":
        (
            len(
                cell60_attribution_state_type_counts
            )
            >
            0

            and

            len(
                cell60_multi_part_attribution_examples
            )
            >
            0
        ),

    "actual_target_assembly_source_snippets_found":
        (
            len(
                cell60_assembly_source_snippets
            )
            >
            0
        ),

    "legacy_selection_basis_resolved":
        (
            cell60_legacy_selection_basis_counts
            ==
            Counter(
                {
                    "EXACT_EXPLICIT_BOOK_REGION":
                        38
                }
            )
        ),

    "legacy_citation_status_resolved":
        (
            cell60_legacy_citation_status_counts
            ==
            Counter(
                {
                    "TARGET_LOCATORS_CAPTURED_NOT_YET_JUDGED":
                        38
                }
            )
        ),

    "v2_target_materialization_not_started":
        (
            EXERCISE_V2_TARGET_ASSEMBLY_DISCOVERY_V1_DRAFT[
                "v2_target_materialization_status"
            ]
            ==
            "NOT_STARTED"
        ),

    "case_activation_not_started":
        (
            EXERCISE_V2_TARGET_ASSEMBLY_DISCOVERY_V1_DRAFT[
                "case_activation_status"
            ]
            ==
            "NOT_STARTED"
        ),

    "task_activation_not_started":
        (
            EXERCISE_V2_TARGET_ASSEMBLY_DISCOVERY_V1_DRAFT[
                "task_activation_status"
            ]
            ==
            "NOT_STARTED"
        ),

    "pilot_rebuild_not_started":
        (
            EXERCISE_V2_TARGET_ASSEMBLY_DISCOVERY_V1_DRAFT[
                "pilot_rebuild_status"
            ]
            ==
            "NOT_STARTED"
        ),

    "question_generation_not_started":
        (
            EXERCISE_V2_TARGET_ASSEMBLY_DISCOVERY_V1_DRAFT[
                "question_generation_status"
            ]
            ==
            "NOT_STARTED"
        ),

    "all_active_and_existing_staging_registries_unchanged":
        (
            cell60_target_snapshot_before
            ==
            cell60_target_snapshot_after

            and

            cell60_task_snapshot_before
            ==
            cell60_task_snapshot_after

            and

            cell60_golden_snapshot_before
            ==
            cell60_golden_snapshot_after

            and

            cell60_pilot_snapshot_before
            ==
            cell60_pilot_snapshot_after

            and

            cell60_case_staging_snapshot_before
            ==
            cell60_case_staging_snapshot_after

            and

            cell60_task_staging_snapshot_before
            ==
            cell60_task_staging_snapshot_after

            and

            len(
                golden_case_records
            )
            ==
            724

            and

            len(
                task_draft_plan_rows
            )
            ==
            724

            and

            len(
                target_evidence_records
            )
            ==
            572

            and

            len(
                pilot_selection_rows
            )
            ==
            23
        ),

    "no_artifact_write":
        (
            EXERCISE_V2_TARGET_ASSEMBLY_DISCOVERY_V1_DRAFT[
                "artifact_write_status"
            ]
            ==
            "NONE"
        ),
}


cell_60_pass = all(
    cell60_gates.values()
)


# ============================================================
# 17) Output
# ============================================================

print(
    "=== V2 TARGET RECORD ASSEMBLY CONTRACT DISCOVERY ==="
)


print(
    "\n=== LEGACY FIELD PROFILES ==="
)


for field_name in (
    CELL60_ASSEMBLY_FIELDS
):

    print(
        "\n",
        field_name,
        sep="",
    )

    print(
        cell60_legacy_field_profiles[
            field_name
        ]
    )


print(
    "\n=== MULTI-PART FIELD PROFILES ==="
)


for field_name in (
    CELL60_ASSEMBLY_FIELDS
):

    print(
        "\n",
        field_name,
        sep="",
    )

    print(
        cell60_multi_part_field_profiles[
            field_name
        ]
    )


print(
    "\n=== EXACT PROJECTION DISCOVERY ==="
)


print(
    "Text-span projection resolved:",
    cell60_text_span_projection_resolved
)


print(
    "Text-span projection keys:",
    (
        sorted(
            cell60_text_span_keys
        )
        if cell60_text_span_keys
        else None
    )
)


print(
    "Text-span projection failures:",
    len(
        cell60_text_span_projection_failures
    )
)


print(
    "Locator projection resolved:",
    cell60_locator_projection_resolved
)


print(
    "Locator projection keys:",
    (
        sorted(
            cell60_locator_ref_keys
        )
        if cell60_locator_ref_keys
        else None
    )
)


print(
    "Locator projection failures:",
    len(
        cell60_locator_projection_failures
    )
)


print(
    "\n=== AUTHORITY STATE ==="
)


print(
    "Formula resolved:",
    cell60_authority_state_formula_resolved
)


print(
    "Failures:",
    len(
        cell60_authority_state_failures
    )
)


print(
    "\n=== ATTRIBUTION STATE PROFILE ==="
)


print(
    "Type counts:",
    dict(
        cell60_attribution_state_type_counts
    )
)


print(
    "Unique structural forms:",
    len(
        cell60_attribution_state_structure_counts
    )
)


print(
    "\nLegacy Exercise examples:"
)


for row in (
    cell60_legacy_attribution_examples
):

    print(
        "\n",
        row,
        sep="",
    )


print(
    "\nExisting Multi-Part examples:"
)


for row in (
    cell60_multi_part_attribution_examples
):

    print(
        "\n",
        row,
        sep="",
    )


print(
    "\n=== HELPER CALLABLES ==="
)


if (
    cell60_candidate_helper_callables
):

    for row in (
        cell60_candidate_helper_callables
    ):

        print(
            "-",
            row
        )

else:

    print(
        "NONE"
    )


print(
    "\n=== TARGET ASSEMBLY SOURCE ==="
)


for index, row in enumerate(
    cell60_assembly_source_snippets[
        :15
    ],
    start=1,
):

    print(
        "\nSNIPPET",
        index,
        "| cell:",
        row[
            "cell_index"
        ],
        "| line:",
        row[
            "line_index"
        ],
        "| anchors:",
        row[
            "matched_anchors"
        ],
    )

    print(
        row[
            "snippet"
        ]
    )


print(
    "\n=== STATUS / SELECTION BASIS ==="
)


print(
    "Legacy Selection Basis:",
    dict(
        cell60_legacy_selection_basis_counts
    )
)


print(
    "Global Selection Basis:",
    dict(
        cell60_global_selection_basis_counts
    )
)


print(
    "Legacy Citation Status:",
    dict(
        cell60_legacy_citation_status_counts
    )
)


print(
    "Global Citation Status:",
    dict(
        cell60_global_citation_status_counts
    )
)


print(
    "\n=== RESOLVED SUBCONTRACTS ==="
)


for field_name, status in (
    cell60_resolved_subcontracts.items()
):

    print(
        "-",
        field_name,
        "=>",
        status,
    )


print(
    "\n=== ACTIVE COUNTS ==="
)


print(
    "Golden:",
    len(
        golden_case_records
    )
)


print(
    "Tasks:",
    len(
        task_draft_plan_rows
    )
)


print(
    "Targets:",
    len(
        target_evidence_records
    )
)


print(
    "Pilots:",
    len(
        pilot_selection_rows
    )
)


print(
    "\n=== CODE CELL 60 GATES ==="
)


for gate_name, passed in (
    cell60_gates.items()
):

    print(
        gate_name,
        ":",
        "PASS"
        if passed
        else "FAIL"
    )


print(
    "\nFAILED GATES:",
    [
        gate_name

        for gate_name, passed
        in cell60_gates.items()

        if not passed
    ]
)


print(
    "\nCODE CELL 60 GATE:",
    "PASS"
    if cell_60_pass
    else "FAIL"
)


assert cell_60_pass, (
    "V2 Target Record assembly substructure discovery failed. "
    "Do not materialize V2 Targets. Repair this SAME Code "
    "Cell 60 until span, locator, authority and attribution "
    "assembly contracts are inspectable without any active "
    "registry mutation."
)
```

```python
# Code Cell 61

# ============================================================
# Exercise V2 Gold Target Materialization — STAGING ONLY
#
# Builds:
#
#   staged_exercise_v2_target_evidence_records
#   staged_exercise_v2_target_evidence_by_case_id
#   staged_exercise_v2_target_evidence_by_task_draft_id
#
# Gold boundary:
#   FULL_ADJUDICATED_EXERCISE_UNIT
#
# Representation:
#   ORDERED_REAL_DOCUMENT_PARTS
#
# Part role:
#   EXPLICIT_BOOK_EXERCISE_UNIT
#
# IMPORTANT:
#   This Cell does NOT activate or replace any current Target.
#
# NO active Target mutation.
# NO Case activation.
# NO Task activation.
# NO Task-ready transition.
# NO Pilot rebuild.
# NO question generation.
# NO expected-answer generation.
# NO artifact writes.
# ============================================================


# ============================================================
# 1) Freeze active + previous staging state
# ============================================================

cell61_active_target_snapshot_before = hashlib.sha256(
    stable_json_key(
        target_evidence_records
    ).encode("utf-8")
).hexdigest()


cell61_active_target_case_index_snapshot_before = hashlib.sha256(
    stable_json_key(
        target_evidence_by_case_id
    ).encode("utf-8")
).hexdigest()


cell61_active_target_task_index_snapshot_before = hashlib.sha256(
    stable_json_key(
        target_evidence_by_task_draft_id
    ).encode("utf-8")
).hexdigest()


cell61_active_task_snapshot_before = hashlib.sha256(
    stable_json_key(
        task_draft_plan_rows
    ).encode("utf-8")
).hexdigest()


cell61_active_golden_snapshot_before = hashlib.sha256(
    stable_json_key(
        golden_case_records
    ).encode("utf-8")
).hexdigest()


cell61_active_pilot_snapshot_before = hashlib.sha256(
    stable_json_key(
        pilot_selection_rows
    ).encode("utf-8")
).hexdigest()


cell61_case_staging_snapshot_before = hashlib.sha256(
    stable_json_key(
        staged_exercise_v2_golden_case_records
    ).encode("utf-8")
).hexdigest()


cell61_task_staging_snapshot_before = hashlib.sha256(
    stable_json_key(
        staged_exercise_v2_task_draft_plan_rows
    ).encode("utf-8")
).hexdigest()


# ============================================================
# 2) Exact scoped V2 Target constants
# ============================================================

CELL61_TARGET_SCHEMA_VERSION = (
    TARGET_EVIDENCE_SCHEMA_VERSION
)


CELL61_SOURCE_TYPE = (
    "book"
)


CELL61_STRATUM = (
    "exercise_or_self_assessment"
)


CELL61_EVIDENCE_METHOD = (
    EXERCISE_V2_TASK_EVIDENCE_TARGET_METHOD
)


CELL61_TARGET_SELECTION_BASIS = (
    EXERCISE_V2_TARGET_SELECTION_BASIS
)


CELL61_TARGET_PART_ROLE = (
    EXERCISE_V2_TARGET_PART_ROLE
)


CELL61_CITATION_TARGET_STATUS = (
    "TARGET_LOCATORS_CAPTURED_NOT_YET_JUDGED"
)


CELL61_QUESTION_GENERATION_STATUS = (
    "NOT_STARTED"
)


CELL61_EXPECTED_ANSWER_GENERATION_STATUS = (
    "NOT_STARTED"
)


# ============================================================
# 3) Contract assertions from Cells 57–60
# ============================================================

cell61_prerequisite_failures = []


if (
    cell57_target_id_formula_status
    !=
    "RESOLVED_BY_EXISTING_BUILDER_AND_INDEPENDENT_RECOMPUTATION_38_OF_38"
):

    cell61_prerequisite_failures.append(
        "TARGET_ID_FORMULA_NOT_RESOLVED"
    )


if not (
    cell58_full_unit_strategy_technical_feasibility
):

    cell61_prerequisite_failures.append(
        "FULL_UNIT_TARGET_NOT_TECHNICALLY_FEASIBLE"
    )


if (
    EXERCISE_V2_GOLD_TARGET_SEMANTIC_CONTRACT_V1_DRAFT[
        "status"
    ]
    !=
    "SEMANTICS_FROZEN_TARGET_MATERIALIZATION_NOT_STARTED"
):

    cell61_prerequisite_failures.append(
        "GOLD_TARGET_SEMANTICS_NOT_FROZEN"
    )


if not (
    cell60_text_span_projection_resolved
):

    cell61_prerequisite_failures.append(
        "TEXT_SPAN_PROJECTION_NOT_RESOLVED"
    )


if not (
    cell60_locator_projection_resolved
):

    cell61_prerequisite_failures.append(
        "LOCATOR_PROJECTION_NOT_RESOLVED"
    )


if not (
    cell60_authority_state_formula_resolved
):

    cell61_prerequisite_failures.append(
        "AUTHORITY_STATE_FORMULA_NOT_RESOLVED"
    )


# ============================================================
# 4) Staged V2 Case / Task lookup
# ============================================================

cell61_staged_case_by_id = {
    row[
        "golden_case_id"
    ]:
        row

    for row
    in staged_exercise_v2_golden_case_records
}


cell61_staged_task_by_case_id = {
    row[
        "golden_case_id"
    ]:
        row

    for row
    in staged_exercise_v2_task_draft_plan_rows
}


cell61_case_task_linkage_failures = []


for case_id in sorted(
    cell61_staged_case_by_id.keys()
):

    if (
        case_id
        not in
        cell61_staged_task_by_case_id
    ):

        cell61_case_task_linkage_failures.append(
            {
                "golden_case_id":
                    case_id,

                "reason":
                    "STAGED_TASK_MISSING",
            }
        )


# ============================================================
# 5) Exact Target-Part builder for V2
#
# target_part_from_document accepts:
#
#   document_id
#   char_start=None
#   char_end=None
#   role=None
#
# Explicitly passing the new V2 role prevents accidental reuse
# of the Legacy EXPLICIT_BOOK_REGION semantic label.
# ============================================================

def cell61_build_v2_target_part(
    document_id,
):

    part = (
        target_part_from_document(
            document_id=(
                document_id
            ),
            role=(
                CELL61_TARGET_PART_ROLE
            ),
        )
    )


    return part


# ============================================================
# 6) Build all unique V2 document Parts
# ============================================================

cell61_all_evidence_document_ids = [
    document_id

    for case_row
    in staged_exercise_v2_golden_case_records

    for document_id
    in case_row[
        "evidence_document_ids"
    ]
]


cell61_unique_evidence_document_ids = set(
    cell61_all_evidence_document_ids
)


cell61_document_part_by_id = {}

cell61_document_part_build_failures = []


for document_id in sorted(
    cell61_unique_evidence_document_ids
):

    try:

        part = (
            cell61_build_v2_target_part(
                document_id
            )
        )


    except Exception as exc:

        cell61_document_part_build_failures.append(
            {
                "document_id":
                    document_id,

                "error_type":
                    type(
                        exc
                    ).__name__,

                "error_message":
                    str(
                        exc
                    ),
            }
        )

        continue


    cell61_document_part_by_id[
        document_id
    ] = part


# ============================================================
# 7) Independent Part validation
# ============================================================

cell61_part_validation_failures = []


for document_id, part in (
    cell61_document_part_by_id.items()
):

    reasons = []


    if (
        set(
            part.keys()
        )
        !=
        cell58_legacy_part_schema
    ):

        reasons.append(
            "TARGET_PART_SCHEMA_MISMATCH"
        )


    if (
        part.get(
            "document_id"
        )
        !=
        document_id
    ):

        reasons.append(
            "DOCUMENT_ID_MISMATCH"
        )


    if (
        part.get(
            "role"
        )
        !=
        CELL61_TARGET_PART_ROLE
    ):

        reasons.append(
            "V2_TARGET_PART_ROLE_MISMATCH"
        )


    if (
        document_id
        not in
        corpus_target_index
    ):

        reasons.append(
            "DOCUMENT_NOT_IN_CORPUS_TARGET_INDEX"
        )


    else:

        corpus_record = (
            corpus_target_index[
                document_id
            ]
        )


        text = (
            corpus_record[
                "text"
            ]
        )


        char_start = (
            part.get(
                "char_start"
            )
        )


        char_end = (
            part.get(
                "char_end"
            )
        )


        if not (
            isinstance(
                char_start,
                int,
            )

            and

            isinstance(
                char_end,
                int,
            )

            and

            char_start
            ==
            0

            and

            char_end
            ==
            len(
                text
            )
        ):

            reasons.append(
                "PART_IS_NOT_EXACT_FULL_DOCUMENT_SPAN"
            )


        else:

            part_text = (
                text[
                    char_start:
                    char_end
                ]
            )


            recomputed_sha = hashlib.sha256(
                part_text.encode(
                    "utf-8"
                )
            ).hexdigest()


            if (
                recomputed_sha
                !=
                part.get(
                    "target_text_sha256"
                )
            ):

                reasons.append(
                    "TARGET_TEXT_SHA_MISMATCH"
                )


            if (
                part.get(
                    "char_count"
                )
                !=
                len(
                    part_text
                )
            ):

                reasons.append(
                    "CHAR_COUNT_MISMATCH"
                )


        if (
            stable_json_key(
                part.get(
                    "locator"
                )
            )
            !=
            stable_json_key(
                corpus_record.get(
                    "locator"
                )
            )
        ):

            reasons.append(
                "LOCATOR_MISMATCH"
            )


        if (
            stable_json_key(
                part.get(
                    "attribution"
                )
            )
            !=
            stable_json_key(
                corpus_record.get(
                    "attribution"
                )
            )
        ):

            reasons.append(
                "ATTRIBUTION_MISMATCH"
            )


        if (
            part.get(
                "source_id"
            )
            !=
            corpus_record.get(
                "source_id"
            )
        ):

            reasons.append(
                "SOURCE_ID_MISMATCH"
            )


    if (
        reasons
    ):

        cell61_part_validation_failures.append(
            {
                "document_id":
                    document_id,

                "reasons":
                    reasons,
            }
        )


# ============================================================
# 8) Exact Target substructure builders from Cell 60
# ============================================================

def cell61_target_text_spans_from_parts(
    parts,
):

    return [
        {
            "document_id":
                part[
                    "document_id"
                ],

            "role":
                part[
                    "role"
                ],

            "char_start":
                part[
                    "char_start"
                ],

            "char_end":
                part[
                    "char_end"
                ],

            "target_text_sha256":
                part[
                    "target_text_sha256"
                ],
        }

        for part
        in parts
    ]


def cell61_target_locator_refs_from_parts(
    parts,
):

    return [
        {
            "document_id":
                part[
                    "document_id"
                ],

            "locator":
                cell52_safe_clone(
                    part[
                        "locator"
                    ]
                ),
        }

        for part
        in parts
    ]


def cell61_target_attribution_state_from_parts(
    parts,
):

    return [
        {
            "document_id":
                part[
                    "document_id"
                ],

            "attribution":
                cell52_safe_clone(
                    part[
                        "attribution"
                    ]
                ),

            "speaker_state":
                cell52_safe_clone(
                    part[
                        "speaker_state"
                    ]
                ),
        }

        for part
        in parts
    ]


def cell61_target_authority_state_from_task(
    task_row,
):

    return {
        "task_authority_expectation":
            task_row[
                "authority_expectation"
            ],

        "authority_inference_performed":
            False,
    }


# ============================================================
# 9) Staging registries
# ============================================================

staged_exercise_v2_target_evidence_records = []

staged_exercise_v2_target_evidence_by_case_id = {}

staged_exercise_v2_target_evidence_by_task_draft_id = {}


cell61_target_build_failures = []


# ============================================================
# 10) Materialize 38 V2 Target records in staging
# ============================================================

if not (
    cell61_prerequisite_failures
    or
    cell61_case_task_linkage_failures
    or
    cell61_document_part_build_failures
    or
    cell61_part_validation_failures
):

    for case_row in (
        staged_exercise_v2_golden_case_records
    ):

        case_id = (
            case_row[
                "golden_case_id"
            ]
        )


        task_row = (
            cell61_staged_task_by_case_id[
                case_id
            ]
        )


        try:

            document_ids = list(
                case_row[
                    "evidence_document_ids"
                ]
            )


            parts = [
                cell52_safe_clone(
                    cell61_document_part_by_id[
                        document_id
                    ]
                )

                for document_id
                in document_ids
            ]


            target_evidence_id = (
                build_target_evidence_id(
                    task_draft_id=(
                        task_row[
                            "task_draft_id"
                        ]
                    ),
                    method=(
                        CELL61_EVIDENCE_METHOD
                    ),
                    parts=(
                        parts
                    ),
                )
            )


            target_record = {
                "target_evidence_id":
                    target_evidence_id,

                "target_schema_version":
                    CELL61_TARGET_SCHEMA_VERSION,

                "task_draft_id":
                    task_row[
                        "task_draft_id"
                    ],

                "golden_case_id":
                    case_id,

                "source_type":
                    CELL61_SOURCE_TYPE,

                "stratum":
                    CELL61_STRATUM,

                "evidence_target_method":
                    CELL61_EVIDENCE_METHOD,

                "target_selection_basis":
                    CELL61_TARGET_SELECTION_BASIS,

                "target_parts":
                    parts,

                "target_document_ids":
                    list(
                        document_ids
                    ),

                "target_locator_refs":
                    cell61_target_locator_refs_from_parts(
                        parts
                    ),

                "target_text_spans":
                    cell61_target_text_spans_from_parts(
                        parts
                    ),

                "target_attribution_state":
                    cell61_target_attribution_state_from_parts(
                        parts
                    ),

                "target_authority_state":
                    cell61_target_authority_state_from_task(
                        task_row
                    ),

                "target_total_tokens":
                    sum(
                        part[
                            "token_count"
                        ]

                        for part
                        in parts
                    ),

                "question_generation_status":
                    CELL61_QUESTION_GENERATION_STATUS,

                "expected_answer_generation_status":
                    CELL61_EXPECTED_ANSWER_GENERATION_STATUS,

                "citation_target_status":
                    CELL61_CITATION_TARGET_STATUS,
            }


            if (
                set(
                    target_record.keys()
                )
                !=
                cell57_target_key_union
            ):

                raise AssertionError(
                    "V2 Target field set does not match the "
                    "actual target_evidence_v1_draft schema."
                )


            if (
                case_id
                in
                staged_exercise_v2_target_evidence_by_case_id
            ):

                raise AssertionError(
                    "Duplicate staged V2 Golden Case ID."
                )


            if (
                task_row[
                    "task_draft_id"
                ]
                in
                staged_exercise_v2_target_evidence_by_task_draft_id
            ):

                raise AssertionError(
                    "Duplicate staged V2 Task Draft ID."
                )


            staged_exercise_v2_target_evidence_records.append(
                target_record
            )


            staged_exercise_v2_target_evidence_by_case_id[
                case_id
            ] = (
                target_record
            )


            staged_exercise_v2_target_evidence_by_task_draft_id[
                task_row[
                    "task_draft_id"
                ]
            ] = (
                target_record
            )


        except Exception as exc:

            cell61_target_build_failures.append(
                {
                    "golden_case_id":
                        case_id,

                    "error_type":
                        type(
                            exc
                        ).__name__,

                    "error_message":
                        str(
                            exc
                        ),
                }
            )


staged_exercise_v2_target_evidence_records = sorted(
    staged_exercise_v2_target_evidence_records,
    key=lambda row:
        row[
            "golden_case_id"
        ],
)


# ============================================================
# 11) Exact Case / Task / Target coverage
# ============================================================

cell61_expected_case_ids = set(
    cell61_staged_case_by_id.keys()
)


cell61_staged_target_case_ids = {
    row[
        "golden_case_id"
    ]

    for row
    in staged_exercise_v2_target_evidence_records
}


cell61_expected_task_ids = {
    row[
        "task_draft_id"
    ]

    for row
    in staged_exercise_v2_task_draft_plan_rows
}


cell61_staged_target_task_ids = {
    row[
        "task_draft_id"
    ]

    for row
    in staged_exercise_v2_target_evidence_records
}


# ============================================================
# 12) Independent Target-ID recomputation
# ============================================================

cell61_target_id_recomputation_failures = []


for target_row in (
    staged_exercise_v2_target_evidence_records
):

    payload = {
        "schema_version":
            TARGET_EVIDENCE_SCHEMA_VERSION,

        "task_draft_id":
            target_row[
                "task_draft_id"
            ],

        "method":
            target_row[
                "evidence_target_method"
            ],

        "parts":
            [
                target_part_identity(
                    part
                )

                for part
                in target_row[
                    "target_parts"
                ]
            ],
    }


    digest = hashlib.sha256(
        stable_json_key(
            payload
        ).encode(
            "utf-8"
        )
    ).hexdigest()


    independent_target_id = (
        "target-evidence:"
        +
        digest[
            :32
        ]
    )


    if (
        independent_target_id
        !=
        target_row[
            "target_evidence_id"
        ]
    ):

        cell61_target_id_recomputation_failures.append(
            {
                "golden_case_id":
                    target_row[
                        "golden_case_id"
                    ],

                "stored":
                    target_row[
                        "target_evidence_id"
                    ],

                "recomputed":
                    independent_target_id,
            }
        )


# ============================================================
# 13) Independent nested projection validation
# ============================================================

cell61_projection_failures = []


for target_row in (
    staged_exercise_v2_target_evidence_records
):

    parts = (
        target_row[
            "target_parts"
        ]
    )


    expected_spans = (
        cell61_target_text_spans_from_parts(
            parts
        )
    )


    expected_locator_refs = (
        cell61_target_locator_refs_from_parts(
            parts
        )
    )


    expected_attribution_state = (
        cell61_target_attribution_state_from_parts(
            parts
        )
    )


    task_row = (
        staged_exercise_v2_task_draft_plans[
            target_row[
                "task_draft_id"
            ]
        ]
    )


    expected_authority_state = (
        cell61_target_authority_state_from_task(
            task_row
        )
    )


    if (
        stable_json_key(
            target_row[
                "target_text_spans"
            ]
        )
        !=
        stable_json_key(
            expected_spans
        )
    ):

        cell61_projection_failures.append(
            {
                "golden_case_id":
                    target_row[
                        "golden_case_id"
                    ],

                "field":
                    "target_text_spans",
            }
        )


    if (
        stable_json_key(
            target_row[
                "target_locator_refs"
            ]
        )
        !=
        stable_json_key(
            expected_locator_refs
        )
    ):

        cell61_projection_failures.append(
            {
                "golden_case_id":
                    target_row[
                        "golden_case_id"
                    ],

                "field":
                    "target_locator_refs",
            }
        )


    if (
        stable_json_key(
            target_row[
                "target_attribution_state"
            ]
        )
        !=
        stable_json_key(
            expected_attribution_state
        )
    ):

        cell61_projection_failures.append(
            {
                "golden_case_id":
                    target_row[
                        "golden_case_id"
                    ],

                "field":
                    "target_attribution_state",
            }
        )


    if (
        stable_json_key(
            target_row[
                "target_authority_state"
            ]
        )
        !=
        stable_json_key(
            expected_authority_state
        )
    ):

        cell61_projection_failures.append(
            {
                "golden_case_id":
                    target_row[
                        "golden_case_id"
                    ],

                "field":
                    "target_authority_state",
            }
        )


# ============================================================
# 14) Exercise boundary / document ordering validation
# ============================================================

cell61_boundary_failures = []


for target_row in (
    staged_exercise_v2_target_evidence_records
):

    case_row = (
        cell61_staged_case_by_id[
            target_row[
                "golden_case_id"
            ]
        ]
    )


    expected_document_ids = list(
        case_row[
            "evidence_document_ids"
        ]
    )


    actual_document_ids = list(
        target_row[
            "target_document_ids"
        ]
    )


    part_document_ids = [
        part[
            "document_id"
        ]

        for part
        in target_row[
            "target_parts"
        ]
    ]


    if not (
        expected_document_ids
        ==
        actual_document_ids
        ==
        part_document_ids
    ):

        cell61_boundary_failures.append(
            {
                "golden_case_id":
                    target_row[
                        "golden_case_id"
                    ],

                "reason":
                    "TARGET_DOES_NOT_EXACTLY_MATCH_CASE_EVIDENCE_ORDER",
            }
        )


# ============================================================
# 15) Token reconciliation with frozen Case snapshot
# ============================================================

cell61_token_reconciliation_failures = []


for target_row in (
    staged_exercise_v2_target_evidence_records
):

    case_row = (
        cell61_staged_case_by_id[
            target_row[
                "golden_case_id"
            ]
        ]
    )


    frozen_tokens = (
        case_row[
            "candidate_snapshot"
        ][
            "exercise_core_token_count"
        ]
    )


    target_tokens = (
        target_row[
            "target_total_tokens"
        ]
    )


    recomputed_part_tokens = sum(
        part[
            "token_count"
        ]

        for part
        in target_row[
            "target_parts"
        ]
    )


    if not (
        frozen_tokens
        ==
        target_tokens
        ==
        recomputed_part_tokens
    ):

        cell61_token_reconciliation_failures.append(
            {
                "golden_case_id":
                    target_row[
                        "golden_case_id"
                    ],

                "frozen_tokens":
                    frozen_tokens,

                "stored_target_tokens":
                    target_tokens,

                "recomputed_part_tokens":
                    recomputed_part_tokens,
            }
        )


# ============================================================
# 16) Locator / span integrity against corpus_target_index
# ============================================================

cell61_span_locator_integrity_failures = []


for target_row in (
    staged_exercise_v2_target_evidence_records
):

    for part in (
        target_row[
            "target_parts"
        ]
    ):

        document_id = (
            part[
                "document_id"
            ]
        )


        corpus_record = (
            corpus_target_index[
                document_id
            ]
        )


        text = (
            corpus_record[
                "text"
            ]
        )


        char_start = (
            part[
                "char_start"
            ]
        )


        char_end = (
            part[
                "char_end"
            ]
        )


        if not (
            0
            <=
            char_start
            <
            char_end
            <=
            len(
                text
            )
        ):

            cell61_span_locator_integrity_failures.append(
                {
                    "golden_case_id":
                        target_row[
                            "golden_case_id"
                        ],

                    "document_id":
                        document_id,

                    "reason":
                        "INVALID_SPAN",
                }
            )

            continue


        text_slice = (
            text[
                char_start:
                char_end
            ]
        )


        sha = hashlib.sha256(
            text_slice.encode(
                "utf-8"
            )
        ).hexdigest()


        if (
            sha
            !=
            part[
                "target_text_sha256"
            ]
        ):

            cell61_span_locator_integrity_failures.append(
                {
                    "golden_case_id":
                        target_row[
                            "golden_case_id"
                        ],

                    "document_id":
                        document_id,

                    "reason":
                        "TEXT_SHA_MISMATCH",
                }
            )


        if (
            stable_json_key(
                part[
                    "locator"
                ]
            )
            !=
            stable_json_key(
                corpus_record[
                    "locator"
                ]
            )
        ):

            cell61_span_locator_integrity_failures.append(
                {
                    "golden_case_id":
                        target_row[
                            "golden_case_id"
                        ],

                    "document_id":
                        document_id,

                    "reason":
                        "LOCATOR_MISMATCH",
                }
            )


# ============================================================
# 17) Attribution / authority integrity
#
# No speaker inference.
# No authority promotion.
# ============================================================

cell61_attribution_authority_failures = []


for target_row in (
    staged_exercise_v2_target_evidence_records
):

    task_row = (
        staged_exercise_v2_task_draft_plans[
            target_row[
                "task_draft_id"
            ]
        ]
    )


    if (
        target_row[
            "target_authority_state"
        ][
            "authority_inference_performed"
        ]
        is not False
    ):

        cell61_attribution_authority_failures.append(
            {
                "golden_case_id":
                    target_row[
                        "golden_case_id"
                    ],

                "reason":
                    "AUTHORITY_INFERENCE_WAS_PERFORMED",
            }
        )


    if (
        target_row[
            "target_authority_state"
        ][
            "task_authority_expectation"
        ]
        !=
        task_row[
            "authority_expectation"
        ]
    ):

        cell61_attribution_authority_failures.append(
            {
                "golden_case_id":
                    target_row[
                        "golden_case_id"
                    ],

                "reason":
                    "AUTHORITY_EXPECTATION_MISMATCH",
            }
        )


    for part, attribution_state in zip(
        target_row[
            "target_parts"
        ],
        target_row[
            "target_attribution_state"
        ],
    ):

        if (
            attribution_state[
                "document_id"
            ]
            !=
            part[
                "document_id"
            ]
        ):

            cell61_attribution_authority_failures.append(
                {
                    "golden_case_id":
                        target_row[
                            "golden_case_id"
                        ],

                    "reason":
                        "ATTRIBUTION_DOCUMENT_ID_MISMATCH",
                }
            )


        if (
            stable_json_key(
                attribution_state[
                    "attribution"
                ]
            )
            !=
            stable_json_key(
                part[
                    "attribution"
                ]
            )
        ):

            cell61_attribution_authority_failures.append(
                {
                    "golden_case_id":
                        target_row[
                            "golden_case_id"
                        ],

                    "reason":
                        "ATTRIBUTION_STATE_MUTATED",
                }
            )


        if (
            stable_json_key(
                attribution_state[
                    "speaker_state"
                ]
            )
            !=
            stable_json_key(
                part[
                    "speaker_state"
                ]
            )
        ):

            cell61_attribution_authority_failures.append(
                {
                    "golden_case_id":
                        target_row[
                            "golden_case_id"
                        ],

                    "reason":
                        "SPEAKER_STATE_MUTATED_OR_INFERRED",
                }
            )


# ============================================================
# 18) Schema / semantic constant validation
# ============================================================

cell61_target_semantic_failures = []


for target_row in (
    staged_exercise_v2_target_evidence_records
):

    expected_fields = {
        "target_schema_version":
            CELL61_TARGET_SCHEMA_VERSION,

        "source_type":
            CELL61_SOURCE_TYPE,

        "stratum":
            CELL61_STRATUM,

        "evidence_target_method":
            CELL61_EVIDENCE_METHOD,

        "target_selection_basis":
            CELL61_TARGET_SELECTION_BASIS,

        "question_generation_status":
            CELL61_QUESTION_GENERATION_STATUS,

        "expected_answer_generation_status":
            CELL61_EXPECTED_ANSWER_GENERATION_STATUS,

        "citation_target_status":
            CELL61_CITATION_TARGET_STATUS,
    }


    for field_name, expected_value in (
        expected_fields.items()
    ):

        if (
            target_row[
                field_name
            ]
            !=
            expected_value
        ):

            cell61_target_semantic_failures.append(
                {
                    "golden_case_id":
                        target_row[
                            "golden_case_id"
                        ],

                    "field":
                        field_name,

                    "actual":
                        target_row[
                            field_name
                        ],

                    "expected":
                        expected_value,
                }
            )


    if not all(
        part[
            "role"
        ]
        ==
        CELL61_TARGET_PART_ROLE

        for part
        in target_row[
            "target_parts"
        ]
    ):

        cell61_target_semantic_failures.append(
            {
                "golden_case_id":
                    target_row[
                        "golden_case_id"
                    ],

                "field":
                    "target_parts.role",

                "reason":
                    "NOT_ALL_PARTS_USE_V2_EXERCISE_UNIT_ROLE",
            }
        )


# ============================================================
# 19) ID uniqueness / collision with active 572 Targets
# ============================================================

cell61_staged_target_ids = [
    row[
        "target_evidence_id"
    ]

    for row
    in staged_exercise_v2_target_evidence_records
]


cell61_staged_target_id_counts = Counter(
    cell61_staged_target_ids
)


cell61_duplicate_staged_target_ids = sorted(
    [
        target_id

        for target_id, count
        in cell61_staged_target_id_counts.items()

        if (
            count
            >
            1
        )
    ]
)


cell61_active_target_ids = {
    row[
        "target_evidence_id"
    ]

    for row
    in target_evidence_records
}


cell61_target_id_collisions_with_active = sorted(
    set(
        cell61_staged_target_ids
    )
    &
    cell61_active_target_ids
)


# ============================================================
# 20) Aggregate staged target counts / size profile
# ============================================================

cell61_total_staged_parts = sum(
    len(
        row[
            "target_parts"
        ]
    )

    for row
    in staged_exercise_v2_target_evidence_records
)


cell61_unique_staged_target_document_ids = {
    document_id

    for row
    in staged_exercise_v2_target_evidence_records

    for document_id
    in row[
        "target_document_ids"
    ]
}


cell61_part_count_distribution = Counter(
    len(
        row[
            "target_parts"
        ]
    )

    for row
    in staged_exercise_v2_target_evidence_records
)


cell61_token_values = sorted(
    row[
        "target_total_tokens"
    ]

    for row
    in staged_exercise_v2_target_evidence_records
)


cell61_token_summary = {
    "min":
        (
            min(
                cell61_token_values
            )
            if cell61_token_values
            else None
        ),

    "median":
        cell58_median(
            cell61_token_values
        ),

    "p75":
        cell58_nearest_rank_percentile(
            cell61_token_values,
            75,
            100,
        ),

    "p90":
        cell58_nearest_rank_percentile(
            cell61_token_values,
            90,
            100,
        ),

    "max":
        (
            max(
                cell61_token_values
            )
            if cell61_token_values
            else None
        ),
}


# ============================================================
# 21) Staging snapshot
# ============================================================

cell61_target_staging_snapshot_payload = {
    "records":
        staged_exercise_v2_target_evidence_records,

    "by_case_id":
        staged_exercise_v2_target_evidence_by_case_id,

    "by_task_draft_id":
        staged_exercise_v2_target_evidence_by_task_draft_id,
}


cell61_target_staging_snapshot_sha256 = hashlib.sha256(
    stable_json_key(
        cell61_target_staging_snapshot_payload
    ).encode(
        "utf-8"
    )
).hexdigest()


# ============================================================
# 22) Target staging contract
# ============================================================

EXERCISE_V2_TARGET_STAGING_V1_DRAFT = {
    "version":
        "exercise_v2_target_staging_v1_draft",

    "status":
        (
            "STAGED_PENDING_INDEPENDENT_POST_TARGET_TRANSITION"
            if (
                len(
                    staged_exercise_v2_target_evidence_records
                )
                ==
                38
            )
            else
            "STAGING_INCOMPLETE"
        ),

    "target_schema_version":
        CELL61_TARGET_SCHEMA_VERSION,

    "gold_boundary":
        EXERCISE_V2_GOLD_TARGET_BOUNDARY,

    "target_representation":
        EXERCISE_V2_GOLD_TARGET_REPRESENTATION,

    "target_selection_basis":
        CELL61_TARGET_SELECTION_BASIS,

    "evidence_target_method":
        CELL61_EVIDENCE_METHOD,

    "target_part_role":
        CELL61_TARGET_PART_ROLE,

    "target_count":
        len(
            staged_exercise_v2_target_evidence_records
        ),

    "target_part_count":
        cell61_total_staged_parts,

    "unique_evidence_document_count":
        len(
            cell61_unique_staged_target_document_ids
        ),

    "staging_snapshot_sha256":
        cell61_target_staging_snapshot_sha256,

    "task_ready_transition_status":
        "NOT_STARTED",

    "active_target_replacement_status":
        "NOT_STARTED",

    "case_activation_status":
        "NOT_STARTED",

    "task_activation_status":
        "NOT_STARTED",

    "pilot_rebuild_status":
        "NOT_STARTED",

    "question_generation_status":
        "NOT_STARTED",

    "expected_answer_generation_status":
        "NOT_STARTED",

    "artifact_write_status":
        "NONE",
}


# ============================================================
# 23) Active no-mutation recheck
# ============================================================

cell61_active_target_snapshot_after = hashlib.sha256(
    stable_json_key(
        target_evidence_records
    ).encode("utf-8")
).hexdigest()


cell61_active_target_case_index_snapshot_after = hashlib.sha256(
    stable_json_key(
        target_evidence_by_case_id
    ).encode("utf-8")
).hexdigest()


cell61_active_target_task_index_snapshot_after = hashlib.sha256(
    stable_json_key(
        target_evidence_by_task_draft_id
    ).encode("utf-8")
).hexdigest()


cell61_active_task_snapshot_after = hashlib.sha256(
    stable_json_key(
        task_draft_plan_rows
    ).encode("utf-8")
).hexdigest()


cell61_active_golden_snapshot_after = hashlib.sha256(
    stable_json_key(
        golden_case_records
    ).encode("utf-8")
).hexdigest()


cell61_active_pilot_snapshot_after = hashlib.sha256(
    stable_json_key(
        pilot_selection_rows
    ).encode("utf-8")
).hexdigest()


cell61_case_staging_snapshot_after = hashlib.sha256(
    stable_json_key(
        staged_exercise_v2_golden_case_records
    ).encode("utf-8")
).hexdigest()


cell61_task_staging_snapshot_after = hashlib.sha256(
    stable_json_key(
        staged_exercise_v2_task_draft_plan_rows
    ).encode("utf-8")
).hexdigest()


# ============================================================
# 24) Validation gates
# ============================================================

cell61_gates = {
    "all_cells_57_to_60_target_contract_prerequisites_resolved":
        (
            len(
                cell61_prerequisite_failures
            )
            ==
            0
        ),

    "exact_38_staged_v2_cases":
        (
            len(
                staged_exercise_v2_golden_case_records
            )
            ==
            38
        ),

    "exact_38_staged_v2_tasks":
        (
            len(
                staged_exercise_v2_task_draft_plan_rows
            )
            ==
            38
        ),

    "all_v2_cases_have_exactly_one_staged_task":
        (
            len(
                cell61_case_task_linkage_failures
            )
            ==
            0

            and

            set(
                cell61_staged_case_by_id.keys()
            )
            ==
            set(
                cell61_staged_task_by_case_id.keys()
            )
        ),

    "exact_480_unique_exercise_evidence_documents":
        (
            len(
                cell61_unique_evidence_document_ids
            )
            ==
            480

            and

            len(
                cell61_all_evidence_document_ids
            )
            ==
            480
        ),

    "all_480_v2_target_parts_build_successfully":
        (
            len(
                cell61_document_part_by_id
            )
            ==
            480

            and

            len(
                cell61_document_part_build_failures
            )
            ==
            0
        ),

    "all_480_v2_target_parts_pass_independent_integrity_validation":
        (
            len(
                cell61_part_validation_failures
            )
            ==
            0
        ),

    "all_480_parts_use_the_scoped_v2_exercise_unit_role":
        all(
            part[
                "role"
            ]
            ==
            EXERCISE_V2_TARGET_PART_ROLE

            for part
            in cell61_document_part_by_id.values()
        ),

    "exact_38_v2_target_records_staged":
        (
            len(
                staged_exercise_v2_target_evidence_records
            )
            ==
            38

            and

            len(
                staged_exercise_v2_target_evidence_by_case_id
            )
            ==
            38

            and

            len(
                staged_exercise_v2_target_evidence_by_task_draft_id
            )
            ==
            38

            and

            len(
                cell61_target_build_failures
            )
            ==
            0
        ),

    "staged_targets_cover_exactly_the_38_v2_cases":
        (
            cell61_staged_target_case_ids
            ==
            cell61_expected_case_ids
        ),

    "staged_targets_cover_exactly_the_38_v2_tasks":
        (
            cell61_staged_target_task_ids
            ==
            cell61_expected_task_ids
        ),

    "all_38_target_ids_independently_recompute":
        (
            len(
                cell61_target_id_recomputation_failures
            )
            ==
            0
        ),

    "all_nested_target_substructures_exactly_recompute_from_parts_and_task":
        (
            len(
                cell61_projection_failures
            )
            ==
            0
        ),

    "all_targets_exactly_equal_the_frozen_exercise_unit_boundary":
        (
            len(
                cell61_boundary_failures
            )
            ==
            0
        ),

    "all_38_target_token_totals_reconcile_with_frozen_exercise_core_counts":
        (
            len(
                cell61_token_reconciliation_failures
            )
            ==
            0
        ),

    "all_target_spans_hashes_and_locators_are_valid":
        (
            len(
                cell61_span_locator_integrity_failures
            )
            ==
            0
        ),

    "no_attribution_or_authority_inference_is_introduced":
        (
            len(
                cell61_attribution_authority_failures
            )
            ==
            0
        ),

    "all_38_staged_targets_match_frozen_v2_semantics":
        (
            len(
                cell61_target_semantic_failures
            )
            ==
            0
        ),

    "all_38_staged_target_ids_are_unique":
        (
            len(
                cell61_staged_target_ids
            )
            ==
            38

            and

            len(
                set(
                    cell61_staged_target_ids
                )
            )
            ==
            38

            and

            len(
                cell61_duplicate_staged_target_ids
            )
            ==
            0
        ),

    "no_staged_v2_target_id_collides_with_active_572_targets":
        (
            len(
                cell61_target_id_collisions_with_active
            )
            ==
            0
        ),

    "staged_targets_contain_exactly_480_real_evidence_parts":
        (
            cell61_total_staged_parts
            ==
            480

            and

            len(
                cell61_unique_staged_target_document_ids
            )
            ==
            480
        ),

    "active_572_targets_and_indexes_remain_byte_stable":
        (
            cell61_active_target_snapshot_before
            ==
            cell61_active_target_snapshot_after

            and

            cell61_active_target_case_index_snapshot_before
            ==
            cell61_active_target_case_index_snapshot_after

            and

            cell61_active_target_task_index_snapshot_before
            ==
            cell61_active_target_task_index_snapshot_after

            and

            len(
                target_evidence_records
            )
            ==
            572
        ),

    "active_case_task_pilot_and_existing_staging_state_unchanged":
        (
            cell61_active_task_snapshot_before
            ==
            cell61_active_task_snapshot_after

            and

            cell61_active_golden_snapshot_before
            ==
            cell61_active_golden_snapshot_after

            and

            cell61_active_pilot_snapshot_before
            ==
            cell61_active_pilot_snapshot_after

            and

            cell61_case_staging_snapshot_before
            ==
            cell61_case_staging_snapshot_after

            and

            cell61_task_staging_snapshot_before
            ==
            cell61_task_staging_snapshot_after

            and

            len(
                golden_case_records
            )
            ==
            724

            and

            len(
                task_draft_plan_rows
            )
            ==
            724

            and

            len(
                pilot_selection_rows
            )
            ==
            23
        ),

    "task_ready_transition_not_started":
        (
            EXERCISE_V2_TARGET_STAGING_V1_DRAFT[
                "task_ready_transition_status"
            ]
            ==
            "NOT_STARTED"
        ),

    "active_target_replacement_not_started":
        (
            EXERCISE_V2_TARGET_STAGING_V1_DRAFT[
                "active_target_replacement_status"
            ]
            ==
            "NOT_STARTED"
        ),

    "case_activation_not_started":
        (
            EXERCISE_V2_TARGET_STAGING_V1_DRAFT[
                "case_activation_status"
            ]
            ==
            "NOT_STARTED"
        ),

    "task_activation_not_started":
        (
            EXERCISE_V2_TARGET_STAGING_V1_DRAFT[
                "task_activation_status"
            ]
            ==
            "NOT_STARTED"
        ),

    "pilot_rebuild_not_started":
        (
            EXERCISE_V2_TARGET_STAGING_V1_DRAFT[
                "pilot_rebuild_status"
            ]
            ==
            "NOT_STARTED"
        ),

    "question_generation_not_started":
        (
            EXERCISE_V2_TARGET_STAGING_V1_DRAFT[
                "question_generation_status"
            ]
            ==
            "NOT_STARTED"
        ),

    "expected_answer_generation_not_started":
        (
            EXERCISE_V2_TARGET_STAGING_V1_DRAFT[
                "expected_answer_generation_status"
            ]
            ==
            "NOT_STARTED"
        ),

    "no_artifact_write":
        (
            EXERCISE_V2_TARGET_STAGING_V1_DRAFT[
                "artifact_write_status"
            ]
            ==
            "NONE"
        ),
}


cell_61_pass = all(
    cell61_gates.values()
)


# ============================================================
# 25) Output
# ============================================================

print(
    "=== EXERCISE V2 GOLD TARGET STAGING ==="
)


print(
    "\n=== PREREQUISITES ==="
)


print(
    "Prerequisite failures:",
    cell61_prerequisite_failures
)


print(
    "Case/Task linkage failures:",
    len(
        cell61_case_task_linkage_failures
    )
)


print(
    "\n=== V2 DOCUMENT PART BUILD ==="
)


print(
    "Evidence document occurrences:",
    len(
        cell61_all_evidence_document_ids
    )
)


print(
    "Unique Evidence documents:",
    len(
        cell61_unique_evidence_document_ids
    )
)


print(
    "Parts built:",
    len(
        cell61_document_part_by_id
    )
)


print(
    "Part build failures:",
    len(
        cell61_document_part_build_failures
    )
)


print(
    "Part validation failures:",
    len(
        cell61_part_validation_failures
    )
)


print(
    "Scoped V2 Part role:",
    CELL61_TARGET_PART_ROLE
)


if (
    cell61_document_part_build_failures
):

    print(
        "\nFirst Part build failures:"
    )

    for row in (
        cell61_document_part_build_failures[
            :10
        ]
    ):

        print(
            "-",
            row
        )


if (
    cell61_part_validation_failures
):

    print(
        "\nFirst Part validation failures:"
    )

    for row in (
        cell61_part_validation_failures[
            :10
        ]
    ):

        print(
            "-",
            row
        )


print(
    "\n=== STAGED TARGET COUNTS ==="
)


print(
    "Target records:",
    len(
        staged_exercise_v2_target_evidence_records
    )
)


print(
    "Case index:",
    len(
        staged_exercise_v2_target_evidence_by_case_id
    )
)


print(
    "Task index:",
    len(
        staged_exercise_v2_target_evidence_by_task_draft_id
    )
)


print(
    "Target build failures:",
    len(
        cell61_target_build_failures
    )
)


print(
    "Total Target Parts:",
    cell61_total_staged_parts
)


print(
    "Unique Target Documents:",
    len(
        cell61_unique_staged_target_document_ids
    )
)


print(
    "Part-count distribution:",
    dict(
        sorted(
            cell61_part_count_distribution.items()
        )
    )
)


print(
    "Token summary:",
    cell61_token_summary
)


print(
    "\n=== INDEPENDENT VALIDATION ==="
)


print(
    "Target-ID recomputation failures:",
    len(
        cell61_target_id_recomputation_failures
    )
)


print(
    "Nested projection failures:",
    len(
        cell61_projection_failures
    )
)


print(
    "Boundary failures:",
    len(
        cell61_boundary_failures
    )
)


print(
    "Token reconciliation failures:",
    len(
        cell61_token_reconciliation_failures
    )
)


print(
    "Span / Locator integrity failures:",
    len(
        cell61_span_locator_integrity_failures
    )
)


print(
    "Attribution / Authority failures:",
    len(
        cell61_attribution_authority_failures
    )
)


print(
    "Semantic failures:",
    len(
        cell61_target_semantic_failures
    )
)


print(
    "Duplicate staged Target IDs:",
    len(
        cell61_duplicate_staged_target_ids
    )
)


print(
    "Target-ID collisions with active:",
    len(
        cell61_target_id_collisions_with_active
    )
)


print(
    "\nStaging Snapshot SHA256:",
    cell61_target_staging_snapshot_sha256
)


print(
    "\n=== FIRST 3 STAGED V2 TARGETS ==="
)


for target_row in (
    staged_exercise_v2_target_evidence_records[
        :3
    ]
):

    print(
        "\nTarget:",
        target_row[
            "target_evidence_id"
        ]
    )


    print(
        "Case:",
        target_row[
            "golden_case_id"
        ]
    )


    print(
        "Task:",
        target_row[
            "task_draft_id"
        ]
    )


    print(
        "Method:",
        target_row[
            "evidence_target_method"
        ]
    )


    print(
        "Selection basis:",
        target_row[
            "target_selection_basis"
        ]
    )


    print(
        "Parts:",
        len(
            target_row[
                "target_parts"
            ]
        )
    )


    print(
        "Tokens:",
        target_row[
            "target_total_tokens"
        ]
    )


    print(
        "Part roles:",
        sorted(
            {
                part[
                    "role"
                ]

                for part
                in target_row[
                    "target_parts"
                ]
            }
        )
    )


    print(
        "Citation status:",
        target_row[
            "citation_target_status"
        ]
    )


print(
    "\n=== ACTIVE COUNTS — MUST REMAIN UNCHANGED ==="
)


print(
    "Golden:",
    len(
        golden_case_records
    )
)


print(
    "Tasks:",
    len(
        task_draft_plan_rows
    )
)


print(
    "Targets:",
    len(
        target_evidence_records
    )
)


print(
    "Pilots:",
    len(
        pilot_selection_rows
    )
)


print(
    "\n=== CODE CELL 61 GATES ==="
)


for gate_name, passed in (
    cell61_gates.items()
):

    print(
        gate_name,
        ":",
        "PASS"
        if passed
        else "FAIL"
    )


print(
    "\nFAILED GATES:",
    [
        gate_name

        for gate_name, passed
        in cell61_gates.items()

        if not passed
    ]
)


print(
    "\nCODE CELL 61 GATE:",
    "PASS"
    if cell_61_pass
    else "FAIL"
)


assert cell_61_pass, (
    "Exercise V2 Target staging failed. "
    "Do not transition V2 Tasks to question-ready and do not "
    "replace active Targets. Repair this SAME Code Cell 61 "
    "until all 38 multi-part Gold Targets, all 480 evidence "
    "Parts, IDs, spans, locators, attribution, authority and "
    "frozen Exercise boundaries independently validate."
)
```

```python
# Code Cell 63

# ============================================================
# Exercise V2 Pilot Replacement Contract Discovery
#
# Goal:
#   discover how the single legacy Exercise pilot must be
#   reselected after the V2 Case/Target rebuild.
#
# READ-ONLY.
#
# NO Pilot staging.
# NO Pilot activation/replacement.
# NO active Case/Task/Target mutation.
# NO question generation.
# NO expected-answer generation.
# NO artifact writes.
# ============================================================


# ============================================================
# 1) Snapshot all relevant active + V2 staged registries
# ============================================================

cell63_pilot_snapshot_before = hashlib.sha256(
    stable_json_key(
        pilot_selection_rows
    ).encode("utf-8")
).hexdigest()


cell63_golden_snapshot_before = hashlib.sha256(
    stable_json_key(
        golden_case_records
    ).encode("utf-8")
).hexdigest()


cell63_task_snapshot_before = hashlib.sha256(
    stable_json_key(
        task_draft_plan_rows
    ).encode("utf-8")
).hexdigest()


cell63_target_snapshot_before = hashlib.sha256(
    stable_json_key(
        target_evidence_records
    ).encode("utf-8")
).hexdigest()


cell63_v2_case_snapshot_before = hashlib.sha256(
    stable_json_key(
        staged_exercise_v2_golden_case_records
    ).encode("utf-8")
).hexdigest()


cell63_v2_task_snapshot_before = hashlib.sha256(
    stable_json_key(
        staged_exercise_v2_posttarget_task_draft_plan_rows
    ).encode("utf-8")
).hexdigest()


cell63_v2_target_snapshot_before = hashlib.sha256(
    stable_json_key(
        staged_exercise_v2_target_evidence_records
    ).encode("utf-8")
).hexdigest()


# ============================================================
# 2) Pilot schema discovery
# ============================================================

cell63_pilot_rows = list(
    pilot_selection_rows
)


cell63_pilot_key_sets = [
    set(
        row.keys()
    )
    for row
    in cell63_pilot_rows
]


cell63_pilot_key_union = (
    set().union(
        *cell63_pilot_key_sets
    )
    if cell63_pilot_key_sets
    else set()
)


cell63_pilot_key_intersection = (
    set(
        cell63_pilot_key_sets[
            0
        ]
    )
    if cell63_pilot_key_sets
    else set()
)


for key_set in (
    cell63_pilot_key_sets[
        1:
    ]
):
    cell63_pilot_key_intersection &= (
        key_set
    )


cell63_pilot_schema_uniform = (
    cell63_pilot_key_union
    ==
    cell63_pilot_key_intersection
)


cell63_pilot_field_distinct_counts = {}


for field_name in sorted(
    cell63_pilot_key_union
):
    cell63_pilot_field_distinct_counts[
        field_name
    ] = len(
        {
            stable_json_key(
                row.get(
                    field_name
                )
            )
            for row
            in cell63_pilot_rows
        }
    )


# ============================================================
# 3) Dynamically resolve Pilot Case / Task / Target ID fields
#
# Avoid assuming field names before inspecting actual records.
# ============================================================

cell63_active_case_ids = {
    row[
        "golden_case_id"
    ]
    for row
    in golden_case_records
}


cell63_active_task_ids = {
    row[
        "task_draft_id"
    ]
    for row
    in task_draft_plan_rows
}


cell63_active_target_ids = {
    row[
        "target_evidence_id"
    ]
    for row
    in target_evidence_records
}


def cell63_resolve_reference_field(
    rows,
    candidate_fields,
    valid_ids,
):
    matches = []

    for field_name in candidate_fields:
        values = [
            row.get(
                field_name
            )
            for row
            in rows
        ]

        non_null_values = [
            value
            for value
            in values
            if value is not None
        ]

        if not (
            non_null_values
        ):
            continue

        if all(
            isinstance(
                value,
                str,
            )
            and
            value
            in
            valid_ids

            for value
            in non_null_values
        ):
            matches.append(
                field_name
            )

    if (
        len(
            matches
        )
        ==
        1
    ):
        return (
            matches[
                0
            ],
            matches,
        )

    return (
        None,
        matches,
    )


cell63_case_id_field_candidates = [
    field_name
    for field_name
    in sorted(
        cell63_pilot_key_union
    )
    if (
        "case"
        in
        field_name.lower()
        and
        "id"
        in
        field_name.lower()
    )
]


cell63_task_id_field_candidates = [
    field_name
    for field_name
    in sorted(
        cell63_pilot_key_union
    )
    if (
        "task"
        in
        field_name.lower()
        and
        "id"
        in
        field_name.lower()
    )
]


cell63_target_id_field_candidates = [
    field_name
    for field_name
    in sorted(
        cell63_pilot_key_union
    )
    if (
        "target"
        in
        field_name.lower()
        and
        "id"
        in
        field_name.lower()
    )
]


(
    cell63_pilot_case_id_field,
    cell63_pilot_case_id_field_matches,
) = cell63_resolve_reference_field(
    cell63_pilot_rows,
    cell63_case_id_field_candidates,
    cell63_active_case_ids,
)


(
    cell63_pilot_task_id_field,
    cell63_pilot_task_id_field_matches,
) = cell63_resolve_reference_field(
    cell63_pilot_rows,
    cell63_task_id_field_candidates,
    cell63_active_task_ids,
)


(
    cell63_pilot_target_id_field,
    cell63_pilot_target_id_field_matches,
) = cell63_resolve_reference_field(
    cell63_pilot_rows,
    cell63_target_id_field_candidates,
    cell63_active_target_ids,
)


# ============================================================
# 4) Locate the exact active Exercise Pilot
# ============================================================

cell63_exercise_pilot_rows = [
    row
    for row
    in cell63_pilot_rows
    if (
        row.get(
            "source_type"
        )
        ==
        "book"

        and

        row.get(
            "stratum"
        )
        ==
        "exercise_or_self_assessment"
    )
]


# Fallback only if Pilot rows do not persist source_type/stratum.
if (
    len(
        cell63_exercise_pilot_rows
    )
    ==
    0

    and

    cell63_pilot_case_id_field
    is not None
):
    cell63_exercise_pilot_rows = []

    for pilot_row in (
        cell63_pilot_rows
    ):
        case_id = (
            pilot_row[
                cell63_pilot_case_id_field
            ]
        )

        case_row = (
            golden_case_records_by_id.get(
                case_id
            )
        )

        if (
            case_row is not None

            and

            case_row.get(
                "source_type"
            )
            ==
            "book"

            and

            case_row.get(
                "stratum"
            )
            ==
            "exercise_or_self_assessment"
        ):
            cell63_exercise_pilot_rows.append(
                pilot_row
            )


cell63_legacy_exercise_pilot = (
    cell63_exercise_pilot_rows[
        0
    ]
    if (
        len(
            cell63_exercise_pilot_rows
        )
        ==
        1
    )
    else None
)


# ============================================================
# 5) Resolve old Pilot Case -> Task -> Target
# ============================================================

cell63_legacy_pilot_case_id = None
cell63_legacy_pilot_task_id = None
cell63_legacy_pilot_target_id = None

cell63_legacy_pilot_case = None
cell63_legacy_pilot_task = None
cell63_legacy_pilot_target = None


if (
    cell63_legacy_exercise_pilot
    is not None

    and

    cell63_pilot_case_id_field
    is not None
):
    cell63_legacy_pilot_case_id = (
        cell63_legacy_exercise_pilot[
            cell63_pilot_case_id_field
        ]
    )

    cell63_legacy_pilot_case = (
        golden_case_records_by_id.get(
            cell63_legacy_pilot_case_id
        )
    )


if (
    cell63_legacy_exercise_pilot
    is not None

    and

    cell63_pilot_task_id_field
    is not None
):
    cell63_legacy_pilot_task_id = (
        cell63_legacy_exercise_pilot[
            cell63_pilot_task_id_field
        ]
    )

else:
    if (
        cell63_legacy_pilot_case_id
        is not None
    ):
        matching_tasks = [
            row
            for row
            in task_draft_plan_rows
            if (
                row[
                    "golden_case_id"
                ]
                ==
                cell63_legacy_pilot_case_id
            )
        ]

        if (
            len(
                matching_tasks
            )
            ==
            1
        ):
            cell63_legacy_pilot_task_id = (
                matching_tasks[
                    0
                ][
                    "task_draft_id"
                ]
            )


if (
    cell63_legacy_pilot_task_id
    is not None
):
    cell63_legacy_pilot_task = (
        task_draft_plans.get(
            cell63_legacy_pilot_task_id
        )
    )


if (
    cell63_legacy_exercise_pilot
    is not None

    and

    cell63_pilot_target_id_field
    is not None
):
    cell63_legacy_pilot_target_id = (
        cell63_legacy_exercise_pilot[
            cell63_pilot_target_id_field
        ]
    )

else:
    if (
        cell63_legacy_pilot_case_id
        is not None
        and
        cell63_legacy_pilot_case_id
        in
        target_evidence_by_case_id
    ):
        cell63_legacy_pilot_target_id = (
            target_evidence_by_case_id[
                cell63_legacy_pilot_case_id
            ][
                "target_evidence_id"
            ]
        )


if (
    cell63_legacy_pilot_target_id
    is not None
):
    matching_targets = [
        row
        for row
        in target_evidence_records
        if (
            row[
                "target_evidence_id"
            ]
            ==
            cell63_legacy_pilot_target_id
        )
    ]

    if (
        len(
            matching_targets
        )
        ==
        1
    ):
        cell63_legacy_pilot_target = (
            matching_targets[
                0
            ]
        )


# ============================================================
# 6) Validate exact legacy Exercise Pilot linkage
# ============================================================

cell63_legacy_pilot_linkage_failures = []


if (
    cell63_legacy_pilot_case is None
):
    cell63_legacy_pilot_linkage_failures.append(
        "LEGACY_PILOT_CASE_NOT_RESOLVED"
    )


if (
    cell63_legacy_pilot_task is None
):
    cell63_legacy_pilot_linkage_failures.append(
        "LEGACY_PILOT_TASK_NOT_RESOLVED"
    )


if (
    cell63_legacy_pilot_target is None
):
    cell63_legacy_pilot_linkage_failures.append(
        "LEGACY_PILOT_TARGET_NOT_RESOLVED"
    )


if (
    cell63_legacy_pilot_case is not None
    and
    cell63_legacy_pilot_task is not None
    and
    cell63_legacy_pilot_task[
        "golden_case_id"
    ]
    !=
    cell63_legacy_pilot_case_id
):
    cell63_legacy_pilot_linkage_failures.append(
        "CASE_TASK_LINKAGE_MISMATCH"
    )


if (
    cell63_legacy_pilot_target is not None
    and
    cell63_legacy_pilot_target[
        "golden_case_id"
    ]
    !=
    cell63_legacy_pilot_case_id
):
    cell63_legacy_pilot_linkage_failures.append(
        "CASE_TARGET_LINKAGE_MISMATCH"
    )


# ============================================================
# 7) Pilot distribution across active strata
# ============================================================

cell63_active_pilot_stratum_counts = Counter(
    (
        row.get(
            "source_type"
        ),
        row.get(
            "stratum"
        ),
    )
    for row
    in cell63_pilot_rows
)


cell63_nonexercise_pilot_rows = [
    row
    for row
    in cell63_pilot_rows
    if row is not cell63_legacy_exercise_pilot
]


# ============================================================
# 8) Discover actual Pilot selection source
# ============================================================

CELL63_PILOT_SOURCE_TERMS = [
    "pilot_selection_rows",
    "pilot_rows_by_token_size",
    "pilot",
    "median",
    "target_total_tokens",
    "selection_distance",
]


cell63_pilot_source_snippets = []


for code_cell in (
    cell49_code_cells
):
    source = (
        code_cell[
            "source"
        ]
    )

    source_lower = (
        source.lower()
    )

    # Require actual pilot logic, not incidental mentions.
    if (
        "pilot"
        not in
        source_lower
    ):
        continue

    source_lines = (
        source.splitlines()
    )

    for line_index, line in enumerate(
        source_lines
    ):
        matched_terms = [
            term
            for term
            in CELL63_PILOT_SOURCE_TERMS
            if (
                term.lower()
                in
                line.lower()
            )
        ]

        if not (
            matched_terms
        ):
            continue

        start_index = max(
            0,
            line_index - 18,
        )

        end_index = min(
            len(
                source_lines
            ),
            line_index + 55,
        )

        cell63_pilot_source_snippets.append(
            {
                "cell_index":
                    code_cell[
                        "cell_index"
                    ],

                "line_index":
                    line_index,

                "matched_terms":
                    matched_terms,

                "snippet":
                    "\n".join(
                        source_lines[
                            start_index:
                            end_index
                        ]
                    ),
            }
        )


# ============================================================
# 9) Discover existing Pilot-related runtime objects
# ============================================================

cell63_pilot_runtime_objects = []


for variable_name, value in list(
    globals().items()
):
    if (
        "pilot"
        not in
        variable_name.lower()
    ):
        continue

    if (
        variable_name.startswith(
            "cell63_"
        )
    ):
        continue

    cell63_pilot_runtime_objects.append(
        {
            "name":
                variable_name,

            "type":
                type(
                    value
                ).__name__,

            "length":
                (
                    len(
                        value
                    )
                    if isinstance(
                        value,
                        (
                            list,
                            dict,
                            tuple,
                            set,
                        ),
                    )
                    else None
                ),
        }
    )


# ============================================================
# 10) Build the exact 38-case V2 Pilot-eligible pool
#
# Eligibility here means:
#   - V2 Case staged
#   - Post-target Task staged
#   - Target staged + independently validated
#   - question not started
#
# No final Pilot choice yet.
# ============================================================

cell63_v2_case_by_id = {
    row[
        "golden_case_id"
    ]:
        row
    for row
    in staged_exercise_v2_golden_case_records
}


cell63_v2_posttarget_task_by_case = {
    row[
        "golden_case_id"
    ]:
        row
    for row
    in staged_exercise_v2_posttarget_task_draft_plan_rows
}


cell63_v2_target_by_case = {
    row[
        "golden_case_id"
    ]:
        row
    for row
    in staged_exercise_v2_target_evidence_records
}


cell63_v2_pilot_candidate_rows = []

cell63_v2_pilot_candidate_failures = []


for case_id in sorted(
    cell63_v2_case_by_id.keys()
):
    case_row = (
        cell63_v2_case_by_id[
            case_id
        ]
    )

    task_row = (
        cell63_v2_posttarget_task_by_case.get(
            case_id
        )
    )

    target_row = (
        cell63_v2_target_by_case.get(
            case_id
        )
    )


    if (
        task_row is None
    ):
        cell63_v2_pilot_candidate_failures.append(
            {
                "golden_case_id":
                    case_id,
                "reason":
                    "POSTTARGET_TASK_MISSING",
            }
        )
        continue


    if (
        target_row is None
    ):
        cell63_v2_pilot_candidate_failures.append(
            {
                "golden_case_id":
                    case_id,
                "reason":
                    "STAGED_TARGET_MISSING",
            }
        )
        continue


    if (
        task_row[
            "construction_eligibility"
        ]
        !=
        "READY_FOR_QUESTION_DRAFT"
    ):
        cell63_v2_pilot_candidate_failures.append(
            {
                "golden_case_id":
                    case_id,
                "reason":
                    "TASK_NOT_READY_FOR_QUESTION_DRAFT",
            }
        )
        continue


    if (
        task_row[
            "target_evidence_status"
        ]
        !=
        "SELECTED_AND_VALIDATED"
    ):
        cell63_v2_pilot_candidate_failures.append(
            {
                "golden_case_id":
                    case_id,
                "reason":
                    "TARGET_NOT_SELECTED_AND_VALIDATED",
            }
        )
        continue


    if (
        task_row[
            "question_text_status"
        ]
        !=
        "NOT_STARTED"
    ):
        cell63_v2_pilot_candidate_failures.append(
            {
                "golden_case_id":
                    case_id,
                "reason":
                    "QUESTION_ALREADY_STARTED",
            }
        )
        continue


    cell63_v2_pilot_candidate_rows.append(
        {
            "golden_case_id":
                case_id,

            "task_draft_id":
                task_row[
                    "task_draft_id"
                ],

            "target_evidence_id":
                target_row[
                    "target_evidence_id"
                ],

            "population_unit_id":
                case_row[
                    "candidate_identity"
                ][
                    "population_unit_id"
                ],

            "heading_path":
                cell52_safe_clone(
                    case_row[
                        "candidate_snapshot"
                    ][
                        "heading_path"
                    ]
                ),

            "target_total_tokens":
                target_row[
                    "target_total_tokens"
                ],

            "target_part_count":
                len(
                    target_row[
                        "target_parts"
                    ]
                ),
        }
    )


# ============================================================
# 11) Empirical V2 target-size center
#
# This does NOT yet claim that median-distance is the final
# replacement rule. It only computes the same observable
# used by the original Pilot analysis.
# ============================================================

cell63_v2_target_token_values = sorted(
    row[
        "target_total_tokens"
    ]
    for row
    in cell63_v2_pilot_candidate_rows
)


cell63_v2_target_token_median = (
    cell58_median(
        cell63_v2_target_token_values
    )
)


for row in (
    cell63_v2_pilot_candidate_rows
):
    row[
        "distance_to_v2_target_token_median"
    ] = abs(
        row[
            "target_total_tokens"
        ]
        -
        cell63_v2_target_token_median
    )


cell63_v2_candidates_by_median_distance = sorted(
    cell63_v2_pilot_candidate_rows,
    key=lambda row:
        (
            row[
                "distance_to_v2_target_token_median"
            ],
            row[
                "target_total_tokens"
            ],
            row[
                "target_part_count"
            ],
            row[
                "golden_case_id"
            ],
        ),
)


cell63_minimum_median_distance = (
    cell63_v2_candidates_by_median_distance[
        0
    ][
        "distance_to_v2_target_token_median"
    ]
    if cell63_v2_candidates_by_median_distance
    else None
)


cell63_minimum_distance_candidates = [
    row
    for row
    in cell63_v2_candidates_by_median_distance
    if (
        row[
            "distance_to_v2_target_token_median"
        ]
        ==
        cell63_minimum_median_distance
    )
]


# ============================================================
# 12) Legacy Pilot size comparison
# ============================================================

cell63_legacy_pilot_target_tokens = (
    cell63_legacy_pilot_target[
        "target_total_tokens"
    ]
    if (
        cell63_legacy_pilot_target
        is not None
    )
    else None
)


cell63_legacy_pilot_target_part_count = (
    len(
        cell63_legacy_pilot_target[
            "target_parts"
        ]
    )
    if (
        cell63_legacy_pilot_target
        is not None
    )
    else None
)


# ============================================================
# 13) Determine whether selection can already be resolved
#
# We do NOT select the Pilot here.
#
# A unique median-distance candidate is informative, but
# source-level tie-break / selection logic still must be
# respected.
# ============================================================

cell63_pilot_replacement_selection_status = (
    "UNIQUE_MEDIAN_DISTANCE_CANDIDATE_FOUND_"
    "PENDING_SOURCE_RULE_REPRODUCTION"
    if (
        len(
            cell63_minimum_distance_candidates
        )
        ==
        1
    )
    else
    "MULTIPLE_MINIMUM_DISTANCE_CANDIDATES_"
    "TIEBREAK_DISCOVERY_REQUIRED"
)


# ============================================================
# 14) Discovery contract
# ============================================================

EXERCISE_V2_PILOT_REPLACEMENT_DISCOVERY_V1_DRAFT = {
    "version":
        "exercise_v2_pilot_replacement_discovery_v1_draft",

    "status":
        "PILOT_REPLACEMENT_CONTRACT_DISCOVERED_NOT_STAGED",

    "active_pilot_count":
        len(
            cell63_pilot_rows
        ),

    "legacy_exercise_pilot_count":
        len(
            cell63_exercise_pilot_rows
        ),

    "legacy_exercise_case_id":
        cell63_legacy_pilot_case_id,

    "legacy_exercise_task_id":
        cell63_legacy_pilot_task_id,

    "legacy_exercise_target_id":
        cell63_legacy_pilot_target_id,

    "legacy_target_tokens":
        cell63_legacy_pilot_target_tokens,

    "legacy_target_part_count":
        cell63_legacy_pilot_target_part_count,

    "v2_candidate_count":
        len(
            cell63_v2_pilot_candidate_rows
        ),

    "v2_target_token_median":
        cell63_v2_target_token_median,

    "minimum_median_distance":
        cell63_minimum_median_distance,

    "minimum_distance_candidate_count":
        len(
            cell63_minimum_distance_candidates
        ),

    "selection_status":
        cell63_pilot_replacement_selection_status,

    "pilot_replacement_staging_status":
        "NOT_STARTED",

    "active_pilot_replacement_status":
        "NOT_STARTED",

    "case_activation_status":
        "NOT_STARTED",

    "task_activation_status":
        "NOT_STARTED",

    "target_activation_status":
        "NOT_STARTED",

    "question_generation_status":
        "NOT_STARTED",

    "artifact_write_status":
        "NONE",
}


# ============================================================
# 15) No-mutation recheck
# ============================================================

cell63_pilot_snapshot_after = hashlib.sha256(
    stable_json_key(
        pilot_selection_rows
    ).encode("utf-8")
).hexdigest()


cell63_golden_snapshot_after = hashlib.sha256(
    stable_json_key(
        golden_case_records
    ).encode("utf-8")
).hexdigest()


cell63_task_snapshot_after = hashlib.sha256(
    stable_json_key(
        task_draft_plan_rows
    ).encode("utf-8")
).hexdigest()


cell63_target_snapshot_after = hashlib.sha256(
    stable_json_key(
        target_evidence_records
    ).encode("utf-8")
).hexdigest()


cell63_v2_case_snapshot_after = hashlib.sha256(
    stable_json_key(
        staged_exercise_v2_golden_case_records
    ).encode("utf-8")
).hexdigest()


cell63_v2_task_snapshot_after = hashlib.sha256(
    stable_json_key(
        staged_exercise_v2_posttarget_task_draft_plan_rows
    ).encode("utf-8")
).hexdigest()


cell63_v2_target_snapshot_after = hashlib.sha256(
    stable_json_key(
        staged_exercise_v2_target_evidence_records
    ).encode("utf-8")
).hexdigest()


# ============================================================
# 16) Gates
# ============================================================

cell63_gates = {
    "active_pilot_count_is_exactly_23":
        (
            len(
                cell63_pilot_rows
            )
            ==
            23
        ),

    "pilot_schema_is_uniform":
        cell63_pilot_schema_uniform,

    "pilot_case_reference_field_resolved_without_guessing":
        (
            cell63_pilot_case_id_field
            is not None
        ),

    "exactly_one_active_exercise_pilot_exists":
        (
            len(
                cell63_exercise_pilot_rows
            )
            ==
            1
        ),

    "exactly_22_nonexercise_pilots_remain_outside_repair_scope":
        (
            len(
                cell63_nonexercise_pilot_rows
            )
            ==
            22
        ),

    "legacy_exercise_pilot_case_task_target_linkage_resolved":
        (
            len(
                cell63_legacy_pilot_linkage_failures
            )
            ==
            0
        ),

    "legacy_exercise_pilot_is_record_level_target":
        (
            cell63_legacy_pilot_target_part_count
            ==
            1
        ),

    "actual_pilot_selection_source_was_found":
        (
            len(
                cell63_pilot_source_snippets
            )
            >
            0
        ),

    "exact_38_v2_posttarget_pilot_candidates_available":
        (
            len(
                cell63_v2_pilot_candidate_rows
            )
            ==
            38

            and

            len(
                cell63_v2_pilot_candidate_failures
            )
            ==
            0
        ),

    "all_38_v2_candidates_have_validated_target_and_ready_task":
        all(
            (
                row[
                    "golden_case_id"
                ]
                in
                cell63_v2_posttarget_task_by_case

                and

                row[
                    "golden_case_id"
                ]
                in
                cell63_v2_target_by_case
            )
            for row
            in cell63_v2_pilot_candidate_rows
        ),

    "v2_target_token_median_is_computable":
        (
            cell63_v2_target_token_median
            is not None
        ),

    "minimum_distance_candidate_set_is_nonempty":
        (
            len(
                cell63_minimum_distance_candidates
            )
            >=
            1
        ),

    "pilot_selection_is_discovered_but_not_applied":
        (
            EXERCISE_V2_PILOT_REPLACEMENT_DISCOVERY_V1_DRAFT[
                "pilot_replacement_staging_status"
            ]
            ==
            "NOT_STARTED"

            and

            EXERCISE_V2_PILOT_REPLACEMENT_DISCOVERY_V1_DRAFT[
                "active_pilot_replacement_status"
            ]
            ==
            "NOT_STARTED"
        ),

    "all_active_and_v2_staging_registries_remain_unchanged":
        (
            cell63_pilot_snapshot_before
            ==
            cell63_pilot_snapshot_after

            and

            cell63_golden_snapshot_before
            ==
            cell63_golden_snapshot_after

            and

            cell63_task_snapshot_before
            ==
            cell63_task_snapshot_after

            and

            cell63_target_snapshot_before
            ==
            cell63_target_snapshot_after

            and

            cell63_v2_case_snapshot_before
            ==
            cell63_v2_case_snapshot_after

            and

            cell63_v2_task_snapshot_before
            ==
            cell63_v2_task_snapshot_after

            and

            cell63_v2_target_snapshot_before
            ==
            cell63_v2_target_snapshot_after

            and

            len(
                golden_case_records
            )
            ==
            724

            and

            len(
                task_draft_plan_rows
            )
            ==
            724

            and

            len(
                target_evidence_records
            )
            ==
            572

            and

            len(
                pilot_selection_rows
            )
            ==
            23
        ),

    "case_activation_not_started":
        (
            EXERCISE_V2_PILOT_REPLACEMENT_DISCOVERY_V1_DRAFT[
                "case_activation_status"
            ]
            ==
            "NOT_STARTED"
        ),

    "task_activation_not_started":
        (
            EXERCISE_V2_PILOT_REPLACEMENT_DISCOVERY_V1_DRAFT[
                "task_activation_status"
            ]
            ==
            "NOT_STARTED"
        ),

    "target_activation_not_started":
        (
            EXERCISE_V2_PILOT_REPLACEMENT_DISCOVERY_V1_DRAFT[
                "target_activation_status"
            ]
            ==
            "NOT_STARTED"
        ),

    "question_generation_not_started":
        (
            EXERCISE_V2_PILOT_REPLACEMENT_DISCOVERY_V1_DRAFT[
                "question_generation_status"
            ]
            ==
            "NOT_STARTED"
        ),

    "no_artifact_write":
        (
            EXERCISE_V2_PILOT_REPLACEMENT_DISCOVERY_V1_DRAFT[
                "artifact_write_status"
            ]
            ==
            "NONE"
        ),
}


cell_63_pass = all(
    cell63_gates.values()
)


# ============================================================
# 17) Output
# ============================================================

print(
    "=== EXERCISE V2 PILOT REPLACEMENT CONTRACT DISCOVERY ==="
)


print(
    "\n=== PILOT SCHEMA ==="
)


print(
    "Active Pilots:",
    len(
        cell63_pilot_rows
    )
)


print(
    "Schema uniform:",
    cell63_pilot_schema_uniform
)


print(
    "Fields:",
    sorted(
        cell63_pilot_key_union
    )
)


print(
    "Resolved Case-ID field:",
    cell63_pilot_case_id_field
)


print(
    "Resolved Task-ID field:",
    cell63_pilot_task_id_field
)


print(
    "Resolved Target-ID field:",
    cell63_pilot_target_id_field
)


print(
    "\n=== LEGACY EXERCISE PILOT ==="
)


print(
    "Exercise Pilot count:",
    len(
        cell63_exercise_pilot_rows
    )
)


print(
    "Legacy Case:",
    cell63_legacy_pilot_case_id
)


print(
    "Legacy Task:",
    cell63_legacy_pilot_task_id
)


print(
    "Legacy Target:",
    cell63_legacy_pilot_target_id
)


print(
    "Legacy Target tokens:",
    cell63_legacy_pilot_target_tokens
)


print(
    "Legacy Target Parts:",
    cell63_legacy_pilot_target_part_count
)


print(
    "Linkage failures:",
    cell63_legacy_pilot_linkage_failures
)


print(
    "\n=== ACTIVE PILOT STRATA ==="
)


for key, count in sorted(
    cell63_active_pilot_stratum_counts.items(),
    key=lambda item:
        str(
            item[
                0
            ]
        ),
):
    print(
        "-",
        key,
        ":",
        count,
    )


print(
    "\n=== V2 PILOT CANDIDATE POOL ==="
)


print(
    "Eligible V2 Candidates:",
    len(
        cell63_v2_pilot_candidate_rows
    )
)


print(
    "Candidate failures:",
    len(
        cell63_v2_pilot_candidate_failures
    )
)


print(
    "V2 Target-token median:",
    cell63_v2_target_token_median
)


print(
    "Minimum distance to median:",
    cell63_minimum_median_distance
)


print(
    "Minimum-distance candidate count:",
    len(
        cell63_minimum_distance_candidates
    )
)


print(
    "Selection status:",
    cell63_pilot_replacement_selection_status
)


print(
    "\n=== TOP 10 CANDIDATES BY MEDIAN DISTANCE ==="
)


for row in (
    cell63_v2_candidates_by_median_distance[
        :10
    ]
):
    print(
        "- Case:",
        row[
            "golden_case_id"
        ],
        "| Target:",
        row[
            "target_evidence_id"
        ],
        "| tokens:",
        row[
            "target_total_tokens"
        ],
        "| parts:",
        row[
            "target_part_count"
        ],
        "| median distance:",
        row[
            "distance_to_v2_target_token_median"
        ],
        "| heading:",
        row[
            "heading_path"
        ],
    )


print(
    "\n=== PILOT SOURCE EVIDENCE ==="
)


print(
    "Source snippets found:",
    len(
        cell63_pilot_source_snippets
    )
)


for index, row in enumerate(
    cell63_pilot_source_snippets[
        :12
    ],
    start=1,
):
    print(
        "\nSNIPPET",
        index,
        "| cell:",
        row[
            "cell_index"
        ],
        "| line:",
        row[
            "line_index"
        ],
        "| terms:",
        row[
            "matched_terms"
        ],
    )

    print(
        row[
            "snippet"
        ]
    )


print(
    "\n=== EXISTING PILOT RUNTIME OBJECTS ==="
)


for row in sorted(
    cell63_pilot_runtime_objects,
    key=lambda item:
        item[
            "name"
        ],
):
    print(
        "-",
        row
    )


print(
    "\n=== ACTIVE COUNTS — MUST REMAIN UNCHANGED ==="
)


print(
    "Golden:",
    len(
        golden_case_records
    )
)


print(
    "Tasks:",
    len(
        task_draft_plan_rows
    )
)


print(
    "Targets:",
    len(
        target_evidence_records
    )
)


print(
    "Pilots:",
    len(
        pilot_selection_rows
    )
)


print(
    "\n=== CODE CELL 63 GATES ==="
)


for gate_name, passed in (
    cell63_gates.items()
):
    print(
        gate_name,
        ":",
        "PASS"
        if passed
        else "FAIL"
    )


print(
    "\nFAILED GATES:",
    [
        gate_name
        for gate_name, passed
        in cell63_gates.items()
        if not passed
    ]
)


print(
    "\nCODE CELL 63 GATE:",
    "PASS"
    if cell_63_pass
    else "FAIL"
)


assert cell_63_pass, (
    "Exercise V2 Pilot replacement contract discovery failed. "
    "Do not stage or replace the Pilot and do not activate the "
    "V2 Case/Task/Target slice. Repair this SAME Code Cell 63 "
    "until the single legacy Exercise Pilot, its Case/Task/"
    "Target lineage, the 38 eligible V2 candidates and the "
    "actual Pilot selection source logic are all resolved."
)
```

```python
# Code Cell 64

# ============================================================
# Exact Pilot Rule Reproduction + V2 Exercise Pilot Staging
#
# Source-proven Pilot rule:
#
#   median =
#       empirical_nearest_rank(target_total_tokens, 0.50)
#
#   distance =
#       abs(target_total_tokens - median)
#
#   tie-break =
#       SHA256(
#           stable_json_key(
#               {
#                   source_type,
#                   stratum,
#                   target_evidence_id,
#               }
#           )
#       )
#
#   winner =
#       min(
#           distance_to_median,
#           stable_tie_break,
#       )
#
# First:
#   reproduce all 23 active Pilot selections exactly.
#
# Then:
#   apply the exact same rule to the 38 staged V2 Exercise
#   Targets and build ONE replacement Pilot row in staging.
#
# NO active Pilot replacement.
# NO active Case/Task/Target mutation.
# NO question generation.
# NO expected-answer generation.
# NO artifact writes.
# ============================================================


# ============================================================
# 1) Snapshot active + V2 staging state
# ============================================================

cell64_active_pilot_snapshot_before = hashlib.sha256(
    stable_json_key(
        pilot_selection_rows
    ).encode("utf-8")
).hexdigest()


cell64_active_golden_snapshot_before = hashlib.sha256(
    stable_json_key(
        golden_case_records
    ).encode("utf-8")
).hexdigest()


cell64_active_task_snapshot_before = hashlib.sha256(
    stable_json_key(
        task_draft_plan_rows
    ).encode("utf-8")
).hexdigest()


cell64_active_target_snapshot_before = hashlib.sha256(
    stable_json_key(
        target_evidence_records
    ).encode("utf-8")
).hexdigest()


cell64_v2_case_snapshot_before = hashlib.sha256(
    stable_json_key(
        staged_exercise_v2_golden_case_records
    ).encode("utf-8")
).hexdigest()


cell64_v2_task_snapshot_before = hashlib.sha256(
    stable_json_key(
        staged_exercise_v2_posttarget_task_draft_plan_rows
    ).encode("utf-8")
).hexdigest()


cell64_v2_target_snapshot_before = hashlib.sha256(
    stable_json_key(
        staged_exercise_v2_target_evidence_records
    ).encode("utf-8")
).hexdigest()


# ============================================================
# 2) Exact active Pilot schema
# ============================================================

CELL64_PILOT_SCHEMA_FIELDS = set(
    cell63_pilot_key_union
)


cell64_active_selection_basis_counts = Counter(
    row[
        "selection_basis"
    ]

    for row
    in pilot_selection_rows
)


cell64_active_selection_basis_value = (
    next(
        iter(
            cell64_active_selection_basis_counts.keys()
        )
    )
    if (
        len(
            cell64_active_selection_basis_counts
        )
        ==
        1
    )
    else None
)


# ============================================================
# 3) Rebuild active targets_by_stratum exactly
# ============================================================

cell64_active_targets_by_stratum = defaultdict(
    list
)


for target_record in (
    target_evidence_records
):

    cell64_active_targets_by_stratum[
        (
            target_record[
                "source_type"
            ],
            target_record[
                "stratum"
            ],
        )
    ].append(
        target_record
    )


# ============================================================
# 4) Exact Pilot rule helper
# ============================================================

def cell64_stable_pilot_tie_break(
    source_type,
    stratum,
    target_evidence_id,
):

    return hashlib.sha256(
        stable_json_key(
            {
                "source_type":
                    source_type,

                "stratum":
                    stratum,

                "target_evidence_id":
                    target_evidence_id,
            }
        ).encode(
            "utf-8"
        )
    ).hexdigest()


def cell64_select_pilot_target(
    source_type,
    stratum,
    targets,
):

    if not (
        targets
    ):

        raise AssertionError(
            "Pilot selection requires at least one Target."
        )


    target_token_values = [
        target[
            "target_total_tokens"
        ]

        for target
        in targets
    ]


    stratum_median_tokens = (
        empirical_nearest_rank(
            target_token_values,
            0.50,
        )
    )


    ranked_targets = []


    for target in (
        targets
    ):

        distance_to_median = abs(
            target[
                "target_total_tokens"
            ]
            -
            stratum_median_tokens
        )


        stable_tie_break = (
            cell64_stable_pilot_tie_break(
                source_type=(
                    source_type
                ),
                stratum=(
                    stratum
                ),
                target_evidence_id=(
                    target[
                        "target_evidence_id"
                    ]
                ),
            )
        )


        ranked_targets.append(
            {
                "target":
                    target,

                "distance_to_median":
                    distance_to_median,

                "stable_tie_break":
                    stable_tie_break,
            }
        )


    winner = sorted(
        ranked_targets,
        key=lambda item:
            (
                item[
                    "distance_to_median"
                ],
                item[
                    "stable_tie_break"
                ],
            ),
    )[0]


    return {
        "stratum_case_count":
            len(
                targets
            ),

        "stratum_target_token_median":
            stratum_median_tokens,

        "winner":
            winner,

        "ranked_targets":
            ranked_targets,
    }


# ============================================================
# 5) Reproduce all 23 active Pilot selections
# ============================================================

cell64_active_pilot_reproduction_rows = []

cell64_active_pilot_reproduction_failures = []


for persisted_pilot in (
    pilot_selection_rows
):

    source_type = (
        persisted_pilot[
            "source_type"
        ]
    )


    stratum = (
        persisted_pilot[
            "stratum"
        ]
    )


    targets = (
        cell64_active_targets_by_stratum[
            (
                source_type,
                stratum,
            )
        ]
    )


    try:

        reproduction = (
            cell64_select_pilot_target(
                source_type=(
                    source_type
                ),
                stratum=(
                    stratum
                ),
                targets=(
                    targets
                ),
            )
        )


    except Exception as exc:

        cell64_active_pilot_reproduction_failures.append(
            {
                "source_type":
                    source_type,

                "stratum":
                    stratum,

                "reason":
                    "SELECTION_EXCEPTION",

                "error_type":
                    type(
                        exc
                    ).__name__,

                "error_message":
                    str(
                        exc
                    ),
            }
        )

        continue


    winner = (
        reproduction[
            "winner"
        ]
    )


    selected_target = (
        winner[
            "target"
        ]
    )


    reproduced_pilot = {
        "source_type":
            source_type,

        "stratum":
            stratum,

        "stratum_case_count":
            reproduction[
                "stratum_case_count"
            ],

        "stratum_target_token_median":
            reproduction[
                "stratum_target_token_median"
            ],

        "selected_target_total_tokens":
            selected_target[
                "target_total_tokens"
            ],

        "distance_to_median":
            winner[
                "distance_to_median"
            ],

        "golden_case_id":
            selected_target[
                "golden_case_id"
            ],

        "task_draft_id":
            selected_target[
                "task_draft_id"
            ],

        "target_evidence_id":
            selected_target[
                "target_evidence_id"
            ],

        "selection_tie_break_sha256":
            winner[
                "stable_tie_break"
            ],

        "selection_basis":
            persisted_pilot[
                "selection_basis"
            ],
    }


    mismatched_fields = [
        field_name

        for field_name
        in sorted(
            CELL64_PILOT_SCHEMA_FIELDS
        )

        if (
            stable_json_key(
                reproduced_pilot[
                    field_name
                ]
            )
            !=
            stable_json_key(
                persisted_pilot[
                    field_name
                ]
            )
        )
    ]


    cell64_active_pilot_reproduction_rows.append(
        {
            "source_type":
                source_type,

            "stratum":
                stratum,

            "persisted_target_evidence_id":
                persisted_pilot[
                    "target_evidence_id"
                ],

            "reproduced_target_evidence_id":
                reproduced_pilot[
                    "target_evidence_id"
                ],

            "mismatched_fields":
                mismatched_fields,
        }
    )


    if (
        mismatched_fields
    ):

        cell64_active_pilot_reproduction_failures.append(
            {
                "source_type":
                    source_type,

                "stratum":
                    stratum,

                "persisted":
                    persisted_pilot,

                "reproduced":
                    reproduced_pilot,

                "mismatched_fields":
                    mismatched_fields,
            }
        )


# ============================================================
# 6) Exact active reproduction count
# ============================================================

cell64_active_pilot_exact_reproduction_count = (
    len(
        pilot_selection_rows
    )
    -
    len(
        cell64_active_pilot_reproduction_failures
    )
)


# ============================================================
# 7) Build exact V2 Exercise Target pool
# ============================================================

CELL64_V2_SOURCE_TYPE = (
    "book"
)


CELL64_V2_STRATUM = (
    "exercise_or_self_assessment"
)


cell64_v2_exercise_targets = [
    row

    for row
    in staged_exercise_v2_target_evidence_records

    if (
        row[
            "source_type"
        ]
        ==
        CELL64_V2_SOURCE_TYPE

        and

        row[
            "stratum"
        ]
        ==
        CELL64_V2_STRATUM
    )
]


# ============================================================
# 8) Apply exact source-proven Pilot rule to 38 V2 Targets
# ============================================================

cell64_v2_selection_exception = None

cell64_v2_selection_result = None


try:

    cell64_v2_selection_result = (
        cell64_select_pilot_target(
            source_type=(
                CELL64_V2_SOURCE_TYPE
            ),
            stratum=(
                CELL64_V2_STRATUM
            ),
            targets=(
                cell64_v2_exercise_targets
            ),
        )
    )


except Exception as exc:

    cell64_v2_selection_exception = {
        "error_type":
            type(
                exc
            ).__name__,

        "error_message":
            str(
                exc
            ),
    }


# ============================================================
# 9) Exact V2 winner
# ============================================================

cell64_v2_winner = (
    cell64_v2_selection_result[
        "winner"
    ]
    if (
        cell64_v2_selection_result
        is not None
    )
    else None
)


cell64_v2_selected_target = (
    cell64_v2_winner[
        "target"
    ]
    if (
        cell64_v2_winner
        is not None
    )
    else None
)


# ============================================================
# 10) Ranked V2 diagnostics
# ============================================================

cell64_v2_ranked_targets = (
    sorted(
        cell64_v2_selection_result[
            "ranked_targets"
        ],
        key=lambda item:
            (
                item[
                    "distance_to_median"
                ],
                item[
                    "stable_tie_break"
                ],
            ),
    )
    if (
        cell64_v2_selection_result
        is not None
    )
    else []
)


# ============================================================
# 11) Verify selected Case/Task/Target all resolve in staged
# V2 post-target chain
# ============================================================

cell64_v2_winner_linkage_failures = []


if (
    cell64_v2_selected_target
    is not None
):

    selected_case_id = (
        cell64_v2_selected_target[
            "golden_case_id"
        ]
    )


    selected_task_id = (
        cell64_v2_selected_target[
            "task_draft_id"
        ]
    )


    selected_target_id = (
        cell64_v2_selected_target[
            "target_evidence_id"
        ]
    )


    if (
        selected_case_id
        not in
        cell63_v2_case_by_id
    ):

        cell64_v2_winner_linkage_failures.append(
            "SELECTED_CASE_NOT_IN_V2_CASE_STAGING"
        )


    if (
        selected_case_id
        not in
        cell63_v2_posttarget_task_by_case
    ):

        cell64_v2_winner_linkage_failures.append(
            "SELECTED_CASE_NOT_IN_POSTTARGET_TASK_STAGING"
        )


    else:

        staged_task = (
            cell63_v2_posttarget_task_by_case[
                selected_case_id
            ]
        )


        if (
            staged_task[
                "task_draft_id"
            ]
            !=
            selected_task_id
        ):

            cell64_v2_winner_linkage_failures.append(
                "SELECTED_TASK_ID_MISMATCH"
            )


        if (
            staged_task[
                "construction_eligibility"
            ]
            !=
            "READY_FOR_QUESTION_DRAFT"
        ):

            cell64_v2_winner_linkage_failures.append(
                "SELECTED_TASK_NOT_READY_FOR_QUESTION_DRAFT"
            )


        if (
            staged_task[
                "target_evidence_status"
            ]
            !=
            "SELECTED_AND_VALIDATED"
        ):

            cell64_v2_winner_linkage_failures.append(
                "SELECTED_TASK_TARGET_NOT_VALIDATED"
            )


    if (
        selected_case_id
        not in
        cell63_v2_target_by_case
    ):

        cell64_v2_winner_linkage_failures.append(
            "SELECTED_TARGET_NOT_IN_V2_TARGET_STAGING"
        )


    else:

        staged_target = (
            cell63_v2_target_by_case[
                selected_case_id
            ]
        )


        if (
            staged_target[
                "target_evidence_id"
            ]
            !=
            selected_target_id
        ):

            cell64_v2_winner_linkage_failures.append(
                "SELECTED_TARGET_ID_MISMATCH"
            )


# ============================================================
# 12) Stage exactly ONE V2 Exercise Pilot row
#
# This is a NEW staging object only.
# ============================================================

staged_exercise_v2_pilot_replacement_row = None


cell64_pilot_build_failure = None


if (
    cell64_active_pilot_exact_reproduction_count
    ==
    23

    and

    cell64_v2_selected_target
    is not None

    and

    len(
        cell64_v2_winner_linkage_failures
    )
    ==
    0

    and

    cell64_active_selection_basis_value
    is not None
):

    try:

        staged_exercise_v2_pilot_replacement_row = {
            "source_type":
                CELL64_V2_SOURCE_TYPE,

            "stratum":
                CELL64_V2_STRATUM,

            "stratum_case_count":
                cell64_v2_selection_result[
                    "stratum_case_count"
                ],

            "stratum_target_token_median":
                cell64_v2_selection_result[
                    "stratum_target_token_median"
                ],

            "selected_target_total_tokens":
                cell64_v2_selected_target[
                    "target_total_tokens"
                ],

            "distance_to_median":
                cell64_v2_winner[
                    "distance_to_median"
                ],

            "golden_case_id":
                cell64_v2_selected_target[
                    "golden_case_id"
                ],

            "task_draft_id":
                cell64_v2_selected_target[
                    "task_draft_id"
                ],

            "target_evidence_id":
                cell64_v2_selected_target[
                    "target_evidence_id"
                ],

            "selection_tie_break_sha256":
                cell64_v2_winner[
                    "stable_tie_break"
                ],

            "selection_basis":
                cell64_active_selection_basis_value,
        }


        if (
            set(
                staged_exercise_v2_pilot_replacement_row.keys()
            )
            !=
            CELL64_PILOT_SCHEMA_FIELDS
        ):

            raise AssertionError(
                "Staged V2 Pilot row does not match the "
                "active Pilot schema exactly."
            )


    except Exception as exc:

        cell64_pilot_build_failure = {
            "error_type":
                type(
                    exc
                ).__name__,

            "error_message":
                str(
                    exc
                ),
        }


# ============================================================
# 13) Independently recompute staged V2 Pilot fields
# ============================================================

cell64_v2_pilot_recomputation_failures = []


if (
    staged_exercise_v2_pilot_replacement_row
    is not None
):

    independent_tie_break = hashlib.sha256(
        stable_json_key(
            {
                "source_type":
                    staged_exercise_v2_pilot_replacement_row[
                        "source_type"
                    ],

                "stratum":
                    staged_exercise_v2_pilot_replacement_row[
                        "stratum"
                    ],

                "target_evidence_id":
                    staged_exercise_v2_pilot_replacement_row[
                        "target_evidence_id"
                    ],
            }
        ).encode(
            "utf-8"
        )
    ).hexdigest()


    if (
        independent_tie_break
        !=
        staged_exercise_v2_pilot_replacement_row[
            "selection_tie_break_sha256"
        ]
    ):

        cell64_v2_pilot_recomputation_failures.append(
            "TIE_BREAK_SHA_MISMATCH"
        )


    independent_distance = abs(
        staged_exercise_v2_pilot_replacement_row[
            "selected_target_total_tokens"
        ]
        -
        staged_exercise_v2_pilot_replacement_row[
            "stratum_target_token_median"
        ]
    )


    if (
        independent_distance
        !=
        staged_exercise_v2_pilot_replacement_row[
            "distance_to_median"
        ]
    ):

        cell64_v2_pilot_recomputation_failures.append(
            "DISTANCE_TO_MEDIAN_MISMATCH"
        )


    independent_median = empirical_nearest_rank(
        [
            target[
                "target_total_tokens"
            ]
            for target
            in cell64_v2_exercise_targets
        ],
        0.50,
    )


    if (
        independent_median
        !=
        staged_exercise_v2_pilot_replacement_row[
            "stratum_target_token_median"
        ]
    ):

        cell64_v2_pilot_recomputation_failures.append(
            "EMPIRICAL_NEAREST_RANK_MEDIAN_MISMATCH"
        )


# ============================================================
# 14) Verify staged winner is exact rank #1
# ============================================================

cell64_v2_rank1_validation_failure = None


if (
    staged_exercise_v2_pilot_replacement_row
    is not None
    and
    cell64_v2_ranked_targets
):

    rank1 = (
        cell64_v2_ranked_targets[
            0
        ]
    )


    if not (
        rank1[
            "target"
        ][
            "target_evidence_id"
        ]
        ==
        staged_exercise_v2_pilot_replacement_row[
            "target_evidence_id"
        ]

        and

        rank1[
            "stable_tie_break"
        ]
        ==
        staged_exercise_v2_pilot_replacement_row[
            "selection_tie_break_sha256"
        ]

        and

        rank1[
            "distance_to_median"
        ]
        ==
        staged_exercise_v2_pilot_replacement_row[
            "distance_to_median"
        ]
    ):

        cell64_v2_rank1_validation_failure = (
            "STAGED_PILOT_IS_NOT_EXACT_SOURCE_RULE_RANK1"
        )


# ============================================================
# 15) Verify exactly 22 active non-Exercise Pilots are outside
# replacement scope
# ============================================================

cell64_nonexercise_pilot_snapshot = hashlib.sha256(
    stable_json_key(
        cell63_nonexercise_pilot_rows
    ).encode(
        "utf-8"
    )
).hexdigest()


# ============================================================
# 16) Staging snapshot
# ============================================================

cell64_pilot_staging_snapshot_sha256 = (
    hashlib.sha256(
        stable_json_key(
            staged_exercise_v2_pilot_replacement_row
        ).encode(
            "utf-8"
        )
    ).hexdigest()

    if (
        staged_exercise_v2_pilot_replacement_row
        is not None
    )

    else None
)


# ============================================================
# 17) Freeze Pilot replacement staging contract
# ============================================================

EXERCISE_V2_PILOT_REPLACEMENT_STAGING_V1_DRAFT = {
    "version":
        "exercise_v2_pilot_replacement_staging_v1_draft",

    "status":
        (
            "EXACT_REPLACEMENT_PILOT_STAGED_NOT_ACTIVATED"
            if (
                staged_exercise_v2_pilot_replacement_row
                is not None

                and

                len(
                    cell64_active_pilot_reproduction_failures
                )
                ==
                0

                and

                len(
                    cell64_v2_pilot_recomputation_failures
                )
                ==
                0

                and

                cell64_v2_rank1_validation_failure
                is None
            )
            else
            "PILOT_REPLACEMENT_STAGING_INCOMPLETE"
        ),

    "source_rule": {
        "median":
            "empirical_nearest_rank(target_total_tokens, 0.50)",

        "primary_sort":
            "absolute_distance_to_median",

        "tie_break":
            "sha256(stable_json({source_type,stratum,target_evidence_id}))",
    },

    "active_rule_reproduction_count":
        cell64_active_pilot_exact_reproduction_count,

    "active_rule_population":
        23,

    "v2_candidate_count":
        len(
            cell64_v2_exercise_targets
        ),

    "v2_empirical_median":
        (
            cell64_v2_selection_result[
                "stratum_target_token_median"
            ]
            if (
                cell64_v2_selection_result
                is not None
            )
            else None
        ),

    "staged_replacement_case_id":
        (
            staged_exercise_v2_pilot_replacement_row[
                "golden_case_id"
            ]
            if (
                staged_exercise_v2_pilot_replacement_row
                is not None
            )
            else None
        ),

    "staged_replacement_task_id":
        (
            staged_exercise_v2_pilot_replacement_row[
                "task_draft_id"
            ]
            if (
                staged_exercise_v2_pilot_replacement_row
                is not None
            )
            else None
        ),

    "staged_replacement_target_id":
        (
            staged_exercise_v2_pilot_replacement_row[
                "target_evidence_id"
            ]
            if (
                staged_exercise_v2_pilot_replacement_row
                is not None
            )
            else None
        ),

    "staging_snapshot_sha256":
        cell64_pilot_staging_snapshot_sha256,

    "active_pilot_replacement_status":
        "NOT_STARTED",

    "case_activation_status":
        "NOT_STARTED",

    "task_activation_status":
        "NOT_STARTED",

    "target_activation_status":
        "NOT_STARTED",

    "question_generation_status":
        "NOT_STARTED",

    "artifact_write_status":
        "NONE",
}


# ============================================================
# 18) No-mutation recheck
# ============================================================

cell64_active_pilot_snapshot_after = hashlib.sha256(
    stable_json_key(
        pilot_selection_rows
    ).encode("utf-8")
).hexdigest()


cell64_active_golden_snapshot_after = hashlib.sha256(
    stable_json_key(
        golden_case_records
    ).encode("utf-8")
).hexdigest()


cell64_active_task_snapshot_after = hashlib.sha256(
    stable_json_key(
        task_draft_plan_rows
    ).encode("utf-8")
).hexdigest()


cell64_active_target_snapshot_after = hashlib.sha256(
    stable_json_key(
        target_evidence_records
    ).encode("utf-8")
).hexdigest()


cell64_v2_case_snapshot_after = hashlib.sha256(
    stable_json_key(
        staged_exercise_v2_golden_case_records
    ).encode("utf-8")
).hexdigest()


cell64_v2_task_snapshot_after = hashlib.sha256(
    stable_json_key(
        staged_exercise_v2_posttarget_task_draft_plan_rows
    ).encode("utf-8")
).hexdigest()


cell64_v2_target_snapshot_after = hashlib.sha256(
    stable_json_key(
        staged_exercise_v2_target_evidence_records
    ).encode("utf-8")
).hexdigest()


# ============================================================
# 19) Gates
# ============================================================

cell64_gates = {
    "active_pilot_schema_matches_discovered_11_field_contract":
        (
            len(
                CELL64_PILOT_SCHEMA_FIELDS
            )
            ==
            11
        ),

    "active_pilot_selection_basis_is_uniform":
        (
            len(
                cell64_active_selection_basis_counts
            )
            ==
            1

            and

            cell64_active_selection_basis_value
            is not None
        ),

    "exact_source_rule_reproduces_all_23_active_pilots":
        (
            cell64_active_pilot_exact_reproduction_count
            ==
            23

            and

            len(
                cell64_active_pilot_reproduction_failures
            )
            ==
            0
        ),

    "exact_38_v2_exercise_targets_enter_replacement_selection":
        (
            len(
                cell64_v2_exercise_targets
            )
            ==
            38
        ),

    "v2_selection_executes_without_exception":
        (
            cell64_v2_selection_exception
            is None

            and

            cell64_v2_selection_result
            is not None
        ),

    "v2_winner_case_task_target_chain_is_valid":
        (
            len(
                cell64_v2_winner_linkage_failures
            )
            ==
            0
        ),

    "exactly_one_v2_exercise_pilot_row_is_staged":
        (
            staged_exercise_v2_pilot_replacement_row
            is not None

            and

            cell64_pilot_build_failure
            is None
        ),

    "staged_v2_pilot_matches_active_pilot_schema_exactly":
        (
            staged_exercise_v2_pilot_replacement_row
            is not None

            and

            set(
                staged_exercise_v2_pilot_replacement_row.keys()
            )
            ==
            CELL64_PILOT_SCHEMA_FIELDS
        ),

    "staged_pilot_empirical_median_distance_and_tiebreak_independently_recompute":
        (
            len(
                cell64_v2_pilot_recomputation_failures
            )
            ==
            0
        ),

    "staged_pilot_is_exact_rank1_under_source_proven_rule":
        (
            cell64_v2_rank1_validation_failure
            is None
        ),

    "exact_22_nonexercise_pilots_remain_outside_repair_scope":
        (
            len(
                cell63_nonexercise_pilot_rows
            )
            ==
            22
        ),

    "active_pilot_registry_remains_exactly_23_and_byte_stable":
        (
            len(
                pilot_selection_rows
            )
            ==
            23

            and

            cell64_active_pilot_snapshot_before
            ==
            cell64_active_pilot_snapshot_after
        ),

    "active_case_task_target_and_v2_staging_registries_unchanged":
        (
            cell64_active_golden_snapshot_before
            ==
            cell64_active_golden_snapshot_after

            and

            cell64_active_task_snapshot_before
            ==
            cell64_active_task_snapshot_after

            and

            cell64_active_target_snapshot_before
            ==
            cell64_active_target_snapshot_after

            and

            cell64_v2_case_snapshot_before
            ==
            cell64_v2_case_snapshot_after

            and

            cell64_v2_task_snapshot_before
            ==
            cell64_v2_task_snapshot_after

            and

            cell64_v2_target_snapshot_before
            ==
            cell64_v2_target_snapshot_after

            and

            len(
                golden_case_records
            )
            ==
            724

            and

            len(
                task_draft_plan_rows
            )
            ==
            724

            and

            len(
                target_evidence_records
            )
            ==
            572
        ),

    "active_pilot_replacement_not_started":
        (
            EXERCISE_V2_PILOT_REPLACEMENT_STAGING_V1_DRAFT[
                "active_pilot_replacement_status"
            ]
            ==
            "NOT_STARTED"
        ),

    "case_activation_not_started":
        (
            EXERCISE_V2_PILOT_REPLACEMENT_STAGING_V1_DRAFT[
                "case_activation_status"
            ]
            ==
            "NOT_STARTED"
        ),

    "task_activation_not_started":
        (
            EXERCISE_V2_PILOT_REPLACEMENT_STAGING_V1_DRAFT[
                "task_activation_status"
            ]
            ==
            "NOT_STARTED"
        ),

    "target_activation_not_started":
        (
            EXERCISE_V2_PILOT_REPLACEMENT_STAGING_V1_DRAFT[
                "target_activation_status"
            ]
            ==
            "NOT_STARTED"
        ),

    "question_generation_not_started":
        (
            EXERCISE_V2_PILOT_REPLACEMENT_STAGING_V1_DRAFT[
                "question_generation_status"
            ]
            ==
            "NOT_STARTED"
        ),

    "no_artifact_write":
        (
            EXERCISE_V2_PILOT_REPLACEMENT_STAGING_V1_DRAFT[
                "artifact_write_status"
            ]
            ==
            "NONE"
        ),
}


cell_64_pass = all(
    cell64_gates.values()
)


# ============================================================
# 20) Output
# ============================================================

print(
    "=== EXACT PILOT RULE REPRODUCTION + V2 REPLACEMENT STAGING ==="
)


print(
    "\n=== ACTIVE RULE REPRODUCTION ==="
)


print(
    "Active Pilots:",
    len(
        pilot_selection_rows
    )
)


print(
    "Selection Basis counts:",
    dict(
        cell64_active_selection_basis_counts
    )
)


print(
    "Exact reproductions:",
    cell64_active_pilot_exact_reproduction_count,
    "/ 23",
)


print(
    "Reproduction failures:",
    len(
        cell64_active_pilot_reproduction_failures
    )
)


if (
    cell64_active_pilot_reproduction_failures
):

    print(
        "\nFirst reproduction failures:"
    )

    for row in (
        cell64_active_pilot_reproduction_failures[
            :10
        ]
    ):

        print(
            "-",
            row
        )


print(
    "\n=== V2 EXACT SOURCE-RULE SELECTION ==="
)


print(
    "Candidate Targets:",
    len(
        cell64_v2_exercise_targets
    )
)


print(
    "Selection exception:",
    cell64_v2_selection_exception
)


if (
    cell64_v2_selection_result
    is not None
):

    print(
        "Empirical nearest-rank median:",
        cell64_v2_selection_result[
            "stratum_target_token_median"
        ]
    )


print(
    "\n=== TOP 10 EXACTLY RANKED V2 CANDIDATES ==="
)


for rank, row in enumerate(
    cell64_v2_ranked_targets[
        :10
    ],
    start=1,
):

    target = (
        row[
            "target"
        ]
    )


    case_row = (
        cell63_v2_case_by_id[
            target[
                "golden_case_id"
            ]
        ]
    )


    print(
        "- rank:",
        rank,
        "| Case:",
        target[
            "golden_case_id"
        ],
        "| Target:",
        target[
            "target_evidence_id"
        ],
        "| tokens:",
        target[
            "target_total_tokens"
        ],
        "| parts:",
        len(
            target[
                "target_parts"
            ]
        ),
        "| distance:",
        row[
            "distance_to_median"
        ],
        "| tie-break:",
        row[
            "stable_tie_break"
        ],
        "| heading:",
        case_row[
            "candidate_snapshot"
        ][
            "heading_path"
        ],
    )


print(
    "\n=== STAGED V2 EXERCISE PILOT ==="
)


if (
    staged_exercise_v2_pilot_replacement_row
    is not None
):

    for field_name in sorted(
        staged_exercise_v2_pilot_replacement_row.keys()
    ):

        print(
            "-",
            field_name,
            "=",
            staged_exercise_v2_pilot_replacement_row[
                field_name
            ],
        )

else:

    print(
        "NONE"
    )


print(
    "\nWinner linkage failures:",
    cell64_v2_winner_linkage_failures
)


print(
    "Pilot recomputation failures:",
    cell64_v2_pilot_recomputation_failures
)


print(
    "Rank-1 validation failure:",
    cell64_v2_rank1_validation_failure
)


print(
    "Pilot staging snapshot SHA256:",
    cell64_pilot_staging_snapshot_sha256
)


print(
    "\n=== ACTIVE COUNTS — MUST REMAIN UNCHANGED ==="
)


print(
    "Golden:",
    len(
        golden_case_records
    )
)


print(
    "Tasks:",
    len(
        task_draft_plan_rows
    )
)


print(
    "Targets:",
    len(
        target_evidence_records
    )
)


print(
    "Pilots:",
    len(
        pilot_selection_rows
    )
)


print(
    "\n=== CODE CELL 64 GATES ==="
)


for gate_name, passed in (
    cell64_gates.items()
):

    print(
        gate_name,
        ":",
        "PASS"
        if passed
        else "FAIL"
    )


print(
    "\nFAILED GATES:",
    [
        gate_name

        for gate_name, passed
        in cell64_gates.items()

        if not passed
    ]
)


print(
    "\nCODE CELL 64 GATE:",
    "PASS"
    if cell_64_pass
    else "FAIL"
)


assert cell_64_pass, (
    "Exact Pilot rule reproduction or V2 Exercise Pilot "
    "staging failed. Do not activate the repaired Exercise "
    "slice. Repair this SAME Code Cell 64 until the original "
    "Pilot rule reproduces 23/23 active Pilots and exactly one "
    "V2 Exercise Pilot is selected and independently validated "
    "under the same empirical-nearest-rank + SHA256 rule."
)
```

```python
# Code Cell 65 — REPAIRED

# ============================================================
# Atomic Activation Simulation + Residual Dependency Audit
#
# REPAIR:
#   Previous implementation compared some list SHA values
#   after the simulated registries had been sorted.
#
#   That made preservation checks order-sensitive even though
#   registry CONTENT could be exactly preserved.
#
# This repair:
#   - compares non-Exercise content by stable identity key;
#   - validates uniqueness separately;
#   - validates exact Legacy removal / V2 admission;
#   - rebuilds core indexes independently;
#   - keeps residual runtime dependencies as a separate
#     classification gate before actual activation.
#
# SIMULATION ONLY.
#
# NO active mutation.
# NO actual activation.
# NO Pilot replacement.
# NO question generation.
# NO expected-answer generation.
# NO artifact writes.
# ============================================================


# ============================================================
# 1) Generic helpers
# ============================================================

def cell65_records(
    value,
):

    if isinstance(
        value,
        dict,
    ):

        return list(
            value.values()
        )


    return list(
        value
    )


def cell65_sha(
    value,
):

    return hashlib.sha256(
        stable_json_key(
            value
        ).encode(
            "utf-8"
        )
    ).hexdigest()


def cell65_clone_rows(
    rows,
):

    return [
        cell52_safe_clone(
            row
        )
        for row
        in rows
    ]


def cell65_keyed_content_snapshot(
    rows,
    key_builder,
):

    keyed_rows = [
        {
            "key":
                key_builder(
                    row
                ),

            "record":
                row,
        }

        for row
        in rows
    ]


    keyed_rows = sorted(
        keyed_rows,
        key=lambda item:
            stable_json_key(
                item[
                    "key"
                ]
            ),
    )


    return cell65_sha(
        keyed_rows
    )


def cell65_duplicate_keys(
    rows,
    key_builder,
):

    counts = Counter(
        stable_json_key(
            key_builder(
                row
            )
        )

        for row
        in rows
    )


    return sorted(
        [
            key

            for key, count
            in counts.items()

            if (
                count
                >
                1
            )
        ]
    )


# ============================================================
# 2) Snapshot all ACTIVE canonical registries
# ============================================================

cell65_active_snapshots_before = {
    "golden_case_records":
        cell65_sha(
            golden_case_records
        ),

    "auto_case_records":
        cell65_sha(
            auto_case_records
        ),

    "case_groundability_precheck":
        cell65_sha(
            case_groundability_precheck
        ),

    "case_task_design":
        cell65_sha(
            case_task_design
        ),

    "task_draft_plan_rows":
        cell65_sha(
            task_draft_plan_rows
        ),

    "target_evidence_records":
        cell65_sha(
            target_evidence_records
        ),

    "pilot_selection_rows":
        cell65_sha(
            pilot_selection_rows
        ),
}


# ============================================================
# 3) Snapshot fully validated V2 staging inputs
# ============================================================

cell65_staging_snapshots_before = {
    "cases":
        cell65_sha(
            staged_exercise_v2_golden_case_records
        ),

    "groundability":
        cell65_sha(
            staged_exercise_v2_posttarget_groundability_precheck
        ),

    "task_design":
        cell65_sha(
            staged_exercise_v2_posttarget_case_task_design
        ),

    "tasks":
        cell65_sha(
            staged_exercise_v2_posttarget_task_draft_plan_rows
        ),

    "targets":
        cell65_sha(
            staged_exercise_v2_target_evidence_records
        ),

    "pilot":
        cell65_sha(
            staged_exercise_v2_pilot_replacement_row
        ),
}


# ============================================================
# 4) Normalize active registry representations
# ============================================================

cell65_active_auto_rows = (
    cell65_records(
        auto_case_records
    )
)


cell65_active_groundability_rows = (
    cell65_records(
        case_groundability_precheck
    )
)


cell65_active_task_design_rows = (
    cell65_records(
        case_task_design
    )
)


# ============================================================
# 5) Resolve exact active Legacy Exercise slice
# ============================================================

cell65_legacy_exercise_cases = [
    row

    for row
    in golden_case_records

    if (
        row[
            "source_type"
        ]
        ==
        "book"

        and

        row[
            "stratum"
        ]
        ==
        "exercise_or_self_assessment"
    )
]


cell65_legacy_exercise_case_ids = {
    row[
        "golden_case_id"
    ]

    for row
    in cell65_legacy_exercise_cases
}


cell65_legacy_exercise_auto_cases = [
    row

    for row
    in cell65_active_auto_rows

    if (
        row[
            "golden_case_id"
        ]
        in
        cell65_legacy_exercise_case_ids
    )
]


cell65_legacy_exercise_groundability = [
    row

    for row
    in cell65_active_groundability_rows

    if (
        row[
            "golden_case_id"
        ]
        in
        cell65_legacy_exercise_case_ids
    )
]


cell65_legacy_exercise_task_design = [
    row

    for row
    in cell65_active_task_design_rows

    if (
        row[
            "golden_case_id"
        ]
        in
        cell65_legacy_exercise_case_ids
    )
]


cell65_legacy_exercise_tasks = [
    row

    for row
    in task_draft_plan_rows

    if (
        row[
            "golden_case_id"
        ]
        in
        cell65_legacy_exercise_case_ids
    )
]


cell65_legacy_exercise_task_ids = {
    row[
        "task_draft_id"
    ]

    for row
    in cell65_legacy_exercise_tasks
}


cell65_legacy_exercise_targets = [
    row

    for row
    in target_evidence_records

    if (
        row[
            "golden_case_id"
        ]
        in
        cell65_legacy_exercise_case_ids
    )
]


cell65_legacy_exercise_target_ids = {
    row[
        "target_evidence_id"
    ]

    for row
    in cell65_legacy_exercise_targets
}


cell65_legacy_exercise_pilots = [
    row

    for row
    in pilot_selection_rows

    if (
        row[
            "source_type"
        ]
        ==
        "book"

        and

        row[
            "stratum"
        ]
        ==
        "exercise_or_self_assessment"
    )
]


# ============================================================
# 6) Resolve exact staged V2 replacement slice
# ============================================================

cell65_v2_cases = (
    cell65_clone_rows(
        staged_exercise_v2_golden_case_records
    )
)


cell65_v2_case_ids = {
    row[
        "golden_case_id"
    ]

    for row
    in cell65_v2_cases
}


cell65_v2_groundability = (
    cell65_clone_rows(
        cell65_records(
            staged_exercise_v2_posttarget_groundability_precheck
        )
    )
)


cell65_v2_task_design = (
    cell65_clone_rows(
        cell65_records(
            staged_exercise_v2_posttarget_case_task_design
        )
    )
)


cell65_v2_tasks = (
    cell65_clone_rows(
        staged_exercise_v2_posttarget_task_draft_plan_rows
    )
)


cell65_v2_task_ids = {
    row[
        "task_draft_id"
    ]

    for row
    in cell65_v2_tasks
}


cell65_v2_targets = (
    cell65_clone_rows(
        staged_exercise_v2_target_evidence_records
    )
)


cell65_v2_target_ids = {
    row[
        "target_evidence_id"
    ]

    for row
    in cell65_v2_targets
}


cell65_v2_pilot = (
    cell52_safe_clone(
        staged_exercise_v2_pilot_replacement_row
    )
)


# ============================================================
# 7) Legacy / V2 ID disjointness
# ============================================================

cell65_case_id_collisions = sorted(
    cell65_legacy_exercise_case_ids
    &
    cell65_v2_case_ids
)


cell65_task_id_collisions = sorted(
    cell65_legacy_exercise_task_ids
    &
    cell65_v2_task_ids
)


cell65_target_id_collisions = sorted(
    cell65_legacy_exercise_target_ids
    &
    cell65_v2_target_ids
)


# ============================================================
# 8) Simulate Golden Case replacement
# ============================================================

cell65_simulated_golden_case_records = (
    [
        cell52_safe_clone(
            row
        )

        for row
        in golden_case_records

        if (
            row[
                "golden_case_id"
            ]
            not in
            cell65_legacy_exercise_case_ids
        )
    ]
    +
    cell65_clone_rows(
        cell65_v2_cases
    )
)


cell65_simulated_golden_case_records_by_id = {
    row[
        "golden_case_id"
    ]:
        row

    for row
    in cell65_simulated_golden_case_records
}


# ============================================================
# 9) Simulate Auto-Golden replacement
# ============================================================

cell65_simulated_auto_case_records = (
    [
        cell52_safe_clone(
            row
        )

        for row
        in cell65_active_auto_rows

        if (
            row[
                "golden_case_id"
            ]
            not in
            cell65_legacy_exercise_case_ids
        )
    ]
    +
    cell65_clone_rows(
        cell65_v2_cases
    )
)


cell65_simulated_auto_case_records_by_id = {
    row[
        "golden_case_id"
    ]:
        row

    for row
    in cell65_simulated_auto_case_records
}


# ============================================================
# 10) Simulate Groundability replacement
# ============================================================

cell65_simulated_groundability_rows = (
    [
        cell52_safe_clone(
            row
        )

        for row
        in cell65_active_groundability_rows

        if (
            row[
                "golden_case_id"
            ]
            not in
            cell65_legacy_exercise_case_ids
        )
    ]
    +
    cell65_clone_rows(
        cell65_v2_groundability
    )
)


cell65_simulated_groundability_by_case_id = {
    row[
        "golden_case_id"
    ]:
        row

    for row
    in cell65_simulated_groundability_rows
}


# ============================================================
# 11) Simulate Task Design replacement
# ============================================================

cell65_simulated_task_design_rows = (
    [
        cell52_safe_clone(
            row
        )

        for row
        in cell65_active_task_design_rows

        if (
            row[
                "golden_case_id"
            ]
            not in
            cell65_legacy_exercise_case_ids
        )
    ]
    +
    cell65_clone_rows(
        cell65_v2_task_design
    )
)


cell65_simulated_task_design_by_case_id = {
    row[
        "golden_case_id"
    ]:
        row

    for row
    in cell65_simulated_task_design_rows
}


# ============================================================
# 12) Simulate Task Draft replacement
# ============================================================

cell65_simulated_task_draft_plan_rows = (
    [
        cell52_safe_clone(
            row
        )

        for row
        in task_draft_plan_rows

        if (
            row[
                "golden_case_id"
            ]
            not in
            cell65_legacy_exercise_case_ids
        )
    ]
    +
    cell65_clone_rows(
        cell65_v2_tasks
    )
)


cell65_simulated_task_draft_plans = {
    row[
        "task_draft_id"
    ]:
        row

    for row
    in cell65_simulated_task_draft_plan_rows
}


# ============================================================
# 13) Simulate Target replacement + indexes
# ============================================================

cell65_simulated_target_evidence_records = (
    [
        cell52_safe_clone(
            row
        )

        for row
        in target_evidence_records

        if (
            row[
                "golden_case_id"
            ]
            not in
            cell65_legacy_exercise_case_ids
        )
    ]
    +
    cell65_clone_rows(
        cell65_v2_targets
    )
)


cell65_simulated_target_evidence_by_case_id = {
    row[
        "golden_case_id"
    ]:
        row

    for row
    in cell65_simulated_target_evidence_records
}


cell65_simulated_target_evidence_by_task_draft_id = {
    row[
        "task_draft_id"
    ]:
        row

    for row
    in cell65_simulated_target_evidence_records
}


# ============================================================
# 14) Simulate Pilot replacement
# ============================================================

cell65_simulated_pilot_selection_rows = (
    [
        cell52_safe_clone(
            row
        )

        for row
        in pilot_selection_rows

        if not (
            row[
                "source_type"
            ]
            ==
            "book"

            and

            row[
                "stratum"
            ]
            ==
            "exercise_or_self_assessment"
        )
    ]
    +
    [
        cell52_safe_clone(
            cell65_v2_pilot
        )
    ]
)


# ============================================================
# 15) Count reconciliation
# ============================================================

cell65_count_reconciliation = {
    "golden_cases": {
        "before":
            len(
                golden_case_records
            ),

        "legacy_removed":
            len(
                cell65_legacy_exercise_cases
            ),

        "v2_added":
            len(
                cell65_v2_cases
            ),

        "simulated_after":
            len(
                cell65_simulated_golden_case_records
            ),
    },

    "auto_cases": {
        "before":
            len(
                cell65_active_auto_rows
            ),

        "legacy_removed":
            len(
                cell65_legacy_exercise_auto_cases
            ),

        "v2_added":
            len(
                cell65_v2_cases
            ),

        "simulated_after":
            len(
                cell65_simulated_auto_case_records
            ),
    },

    "groundability": {
        "before":
            len(
                cell65_active_groundability_rows
            ),

        "legacy_removed":
            len(
                cell65_legacy_exercise_groundability
            ),

        "v2_added":
            len(
                cell65_v2_groundability
            ),

        "simulated_after":
            len(
                cell65_simulated_groundability_rows
            ),
    },

    "task_design": {
        "before":
            len(
                cell65_active_task_design_rows
            ),

        "legacy_removed":
            len(
                cell65_legacy_exercise_task_design
            ),

        "v2_added":
            len(
                cell65_v2_task_design
            ),

        "simulated_after":
            len(
                cell65_simulated_task_design_rows
            ),
    },

    "task_drafts": {
        "before":
            len(
                task_draft_plan_rows
            ),

        "legacy_removed":
            len(
                cell65_legacy_exercise_tasks
            ),

        "v2_added":
            len(
                cell65_v2_tasks
            ),

        "simulated_after":
            len(
                cell65_simulated_task_draft_plan_rows
            ),
    },

    "targets": {
        "before":
            len(
                target_evidence_records
            ),

        "legacy_removed":
            len(
                cell65_legacy_exercise_targets
            ),

        "v2_added":
            len(
                cell65_v2_targets
            ),

        "simulated_after":
            len(
                cell65_simulated_target_evidence_records
            ),
    },

    "pilots": {
        "before":
            len(
                pilot_selection_rows
            ),

        "legacy_removed":
            len(
                cell65_legacy_exercise_pilots
            ),

        "v2_added":
            (
                1
                if cell65_v2_pilot is not None
                else 0
            ),

        "simulated_after":
            len(
                cell65_simulated_pilot_selection_rows
            ),
    },
}


cell65_core_count_failures = [
    {
        "registry":
            registry_name,

        "before":
            counts[
                "before"
            ],

        "after":
            counts[
                "simulated_after"
            ],
    }

    for registry_name, counts
    in cell65_count_reconciliation.items()

    if (
        counts[
            "before"
        ]
        !=
        counts[
            "simulated_after"
        ]
    )
]


# ============================================================
# 16) Exact uniqueness audits
# ============================================================

cell65_duplicate_simulated_case_ids = (
    cell65_duplicate_keys(
        cell65_simulated_golden_case_records,
        lambda row:
            row[
                "golden_case_id"
            ],
    )
)


cell65_duplicate_simulated_auto_case_ids = (
    cell65_duplicate_keys(
        cell65_simulated_auto_case_records,
        lambda row:
            row[
                "golden_case_id"
            ],
    )
)


cell65_duplicate_simulated_groundability_case_ids = (
    cell65_duplicate_keys(
        cell65_simulated_groundability_rows,
        lambda row:
            row[
                "golden_case_id"
            ],
    )
)


cell65_duplicate_simulated_design_case_ids = (
    cell65_duplicate_keys(
        cell65_simulated_task_design_rows,
        lambda row:
            row[
                "golden_case_id"
            ],
    )
)


cell65_duplicate_simulated_task_ids = (
    cell65_duplicate_keys(
        cell65_simulated_task_draft_plan_rows,
        lambda row:
            row[
                "task_draft_id"
            ],
    )
)


cell65_duplicate_simulated_target_ids = (
    cell65_duplicate_keys(
        cell65_simulated_target_evidence_records,
        lambda row:
            row[
                "target_evidence_id"
            ],
    )
)


cell65_duplicate_simulated_pilot_strata = (
    cell65_duplicate_keys(
        cell65_simulated_pilot_selection_rows,
        lambda row:
            {
                "source_type":
                    row[
                        "source_type"
                    ],

                "stratum":
                    row[
                        "stratum"
                    ],
            },
    )
)


# ============================================================
# 17) Exact Legacy-ID removal
# ============================================================

cell65_old_id_removal_failures = []


for case_id in (
    cell65_legacy_exercise_case_ids
):

    if (
        case_id
        in
        cell65_simulated_golden_case_records_by_id
    ):

        cell65_old_id_removal_failures.append(
            {
                "family":
                    "GOLDEN_CASE",

                "id":
                    case_id,
            }
        )


    if (
        case_id
        in
        cell65_simulated_auto_case_records_by_id
    ):

        cell65_old_id_removal_failures.append(
            {
                "family":
                    "AUTO_CASE",

                "id":
                    case_id,
            }
        )


    if (
        case_id
        in
        cell65_simulated_groundability_by_case_id
    ):

        cell65_old_id_removal_failures.append(
            {
                "family":
                    "GROUNDABILITY",

                "id":
                    case_id,
            }
        )


    if (
        case_id
        in
        cell65_simulated_task_design_by_case_id
    ):

        cell65_old_id_removal_failures.append(
            {
                "family":
                    "TASK_DESIGN",

                "id":
                    case_id,
            }
        )


    if (
        case_id
        in
        cell65_simulated_target_evidence_by_case_id
    ):

        cell65_old_id_removal_failures.append(
            {
                "family":
                    "TARGET_CASE_INDEX",

                "id":
                    case_id,
            }
        )


for task_id in (
    cell65_legacy_exercise_task_ids
):

    if (
        task_id
        in
        cell65_simulated_task_draft_plans
    ):

        cell65_old_id_removal_failures.append(
            {
                "family":
                    "TASK",

                "id":
                    task_id,
            }
        )


    if (
        task_id
        in
        cell65_simulated_target_evidence_by_task_draft_id
    ):

        cell65_old_id_removal_failures.append(
            {
                "family":
                    "TARGET_TASK_INDEX",

                "id":
                    task_id,
            }
        )


cell65_simulated_target_id_set = {
    row[
        "target_evidence_id"
    ]

    for row
    in cell65_simulated_target_evidence_records
}


for target_id in (
    cell65_legacy_exercise_target_ids
):

    if (
        target_id
        in
        cell65_simulated_target_id_set
    ):

        cell65_old_id_removal_failures.append(
            {
                "family":
                    "TARGET",

                "id":
                    target_id,
            }
        )


# ============================================================
# 18) Exact V2-ID admission
# ============================================================

cell65_new_id_admission_failures = []


for case_id in (
    cell65_v2_case_ids
):

    required_case_registries = {
        "GOLDEN_CASE":
            cell65_simulated_golden_case_records_by_id,

        "AUTO_CASE":
            cell65_simulated_auto_case_records_by_id,

        "GROUNDABILITY":
            cell65_simulated_groundability_by_case_id,

        "TASK_DESIGN":
            cell65_simulated_task_design_by_case_id,

        "TARGET_CASE_INDEX":
            cell65_simulated_target_evidence_by_case_id,
    }


    for family_name, registry in (
        required_case_registries.items()
    ):

        if (
            case_id
            not in
            registry
        ):

            cell65_new_id_admission_failures.append(
                {
                    "family":
                        family_name,

                    "id":
                        case_id,
                }
            )


for task_id in (
    cell65_v2_task_ids
):

    if (
        task_id
        not in
        cell65_simulated_task_draft_plans
    ):

        cell65_new_id_admission_failures.append(
            {
                "family":
                    "TASK",

                "id":
                    task_id,
            }
        )


    if (
        task_id
        not in
        cell65_simulated_target_evidence_by_task_draft_id
    ):

        cell65_new_id_admission_failures.append(
            {
                "family":
                    "TARGET_TASK_INDEX",

                "id":
                    task_id,
            }
        )


for target_id in (
    cell65_v2_target_ids
):

    if (
        target_id
        not in
        cell65_simulated_target_id_set
    ):

        cell65_new_id_admission_failures.append(
            {
                "family":
                    "TARGET",

                "id":
                    target_id,
            }
        )


# ============================================================
# 19) Simulated Case -> Groundability -> Design -> Task ->
# Target linkage audit
# ============================================================

cell65_linkage_failures = []


for task_row in (
    cell65_simulated_task_draft_plan_rows
):

    case_id = (
        task_row[
            "golden_case_id"
        ]
    )


    if (
        case_id
        not in
        cell65_simulated_golden_case_records_by_id
    ):

        cell65_linkage_failures.append(
            {
                "family":
                    "TASK",

                "id":
                    task_row[
                        "task_draft_id"
                    ],

                "reason":
                    "CASE_MISSING",
            }
        )


for target_row in (
    cell65_simulated_target_evidence_records
):

    case_id = (
        target_row[
            "golden_case_id"
        ]
    )


    task_id = (
        target_row[
            "task_draft_id"
        ]
    )


    if (
        case_id
        not in
        cell65_simulated_golden_case_records_by_id
    ):

        cell65_linkage_failures.append(
            {
                "family":
                    "TARGET",

                "id":
                    target_row[
                        "target_evidence_id"
                    ],

                "reason":
                    "CASE_MISSING",
            }
        )


    if (
        task_id
        not in
        cell65_simulated_task_draft_plans
    ):

        cell65_linkage_failures.append(
            {
                "family":
                    "TARGET",

                "id":
                    target_row[
                        "target_evidence_id"
                    ],

                "reason":
                    "TASK_MISSING",
            }
        )

        continue


    if (
        cell65_simulated_task_draft_plans[
            task_id
        ][
            "golden_case_id"
        ]
        !=
        case_id
    ):

        cell65_linkage_failures.append(
            {
                "family":
                    "TARGET",

                "id":
                    target_row[
                        "target_evidence_id"
                    ],

                "reason":
                    "TASK_CASE_MISMATCH",
            }
        )


for case_id in (
    cell65_v2_case_ids
):

    if (
        case_id
        not in
        cell65_simulated_groundability_by_case_id
    ):

        cell65_linkage_failures.append(
            {
                "family":
                    "V2_CASE",

                "id":
                    case_id,

                "reason":
                    "GROUNDABILITY_MISSING",
            }
        )


    if (
        case_id
        not in
        cell65_simulated_task_design_by_case_id
    ):

        cell65_linkage_failures.append(
            {
                "family":
                    "V2_CASE",

                "id":
                    case_id,

                "reason":
                    "TASK_DESIGN_MISSING",
            }
        )


    if (
        case_id
        not in
        cell65_simulated_target_evidence_by_case_id
    ):

        cell65_linkage_failures.append(
            {
                "family":
                    "V2_CASE",

                "id":
                    case_id,

                "reason":
                    "TARGET_MISSING",
            }
        )


# ============================================================
# 20) Simulated Pilot linkage
# ============================================================

cell65_simulated_exercise_pilots = [
    row

    for row
    in cell65_simulated_pilot_selection_rows

    if (
        row[
            "source_type"
        ]
        ==
        "book"

        and

        row[
            "stratum"
        ]
        ==
        "exercise_or_self_assessment"
    )
]


cell65_pilot_linkage_failures = []


for pilot_row in (
    cell65_simulated_pilot_selection_rows
):

    case_id = (
        pilot_row[
            "golden_case_id"
        ]
    )


    task_id = (
        pilot_row[
            "task_draft_id"
        ]
    )


    target_id = (
        pilot_row[
            "target_evidence_id"
        ]
    )


    if (
        case_id
        not in
        cell65_simulated_golden_case_records_by_id
    ):

        cell65_pilot_linkage_failures.append(
            {
                "stratum":
                    (
                        pilot_row[
                            "source_type"
                        ],
                        pilot_row[
                            "stratum"
                        ],
                    ),

                "reason":
                    "CASE_MISSING",
            }
        )


    if (
        task_id
        not in
        cell65_simulated_task_draft_plans
    ):

        cell65_pilot_linkage_failures.append(
            {
                "stratum":
                    (
                        pilot_row[
                            "source_type"
                        ],
                        pilot_row[
                            "stratum"
                        ],
                    ),

                "reason":
                    "TASK_MISSING",
            }
        )


    if (
        target_id
        not in
        cell65_simulated_target_id_set
    ):

        cell65_pilot_linkage_failures.append(
            {
                "stratum":
                    (
                        pilot_row[
                            "source_type"
                        ],
                        pilot_row[
                            "stratum"
                        ],
                    ),

                "reason":
                    "TARGET_MISSING",
            }
        )


# ============================================================
# 21) ORDER-INDEPENDENT Non-Exercise preservation
#
# This is the principal repair.
# ============================================================

cell65_active_nonexercise_golden = [
    row

    for row
    in golden_case_records

    if (
        row[
            "golden_case_id"
        ]
        not in
        cell65_legacy_exercise_case_ids
    )
]


cell65_simulated_nonexercise_golden = [
    row

    for row
    in cell65_simulated_golden_case_records

    if (
        row[
            "golden_case_id"
        ]
        not in
        cell65_v2_case_ids
    )
]


cell65_active_nonexercise_auto = [
    row

    for row
    in cell65_active_auto_rows

    if (
        row[
            "golden_case_id"
        ]
        not in
        cell65_legacy_exercise_case_ids
    )
]


cell65_simulated_nonexercise_auto = [
    row

    for row
    in cell65_simulated_auto_case_records

    if (
        row[
            "golden_case_id"
        ]
        not in
        cell65_v2_case_ids
    )
]


cell65_active_nonexercise_groundability = [
    row

    for row
    in cell65_active_groundability_rows

    if (
        row[
            "golden_case_id"
        ]
        not in
        cell65_legacy_exercise_case_ids
    )
]


cell65_simulated_nonexercise_groundability = [
    row

    for row
    in cell65_simulated_groundability_rows

    if (
        row[
            "golden_case_id"
        ]
        not in
        cell65_v2_case_ids
    )
]


cell65_active_nonexercise_design = [
    row

    for row
    in cell65_active_task_design_rows

    if (
        row[
            "golden_case_id"
        ]
        not in
        cell65_legacy_exercise_case_ids
    )
]


cell65_simulated_nonexercise_design = [
    row

    for row
    in cell65_simulated_task_design_rows

    if (
        row[
            "golden_case_id"
        ]
        not in
        cell65_v2_case_ids
    )
]


cell65_active_nonexercise_tasks = [
    row

    for row
    in task_draft_plan_rows

    if (
        row[
            "golden_case_id"
        ]
        not in
        cell65_legacy_exercise_case_ids
    )
]


cell65_simulated_nonexercise_tasks = [
    row

    for row
    in cell65_simulated_task_draft_plan_rows

    if (
        row[
            "golden_case_id"
        ]
        not in
        cell65_v2_case_ids
    )
]


cell65_active_nonexercise_targets = [
    row

    for row
    in target_evidence_records

    if (
        row[
            "golden_case_id"
        ]
        not in
        cell65_legacy_exercise_case_ids
    )
]


cell65_simulated_nonexercise_targets = [
    row

    for row
    in cell65_simulated_target_evidence_records

    if (
        row[
            "golden_case_id"
        ]
        not in
        cell65_v2_case_ids
    )
]


cell65_active_nonexercise_pilots = [
    row

    for row
    in pilot_selection_rows

    if not (
        row[
            "source_type"
        ]
        ==
        "book"

        and

        row[
            "stratum"
        ]
        ==
        "exercise_or_self_assessment"
    )
]


cell65_simulated_nonexercise_pilots = [
    row

    for row
    in cell65_simulated_pilot_selection_rows

    if not (
        row[
            "source_type"
        ]
        ==
        "book"

        and

        row[
            "stratum"
        ]
        ==
        "exercise_or_self_assessment"
    )
]


cell65_nonexercise_preservation = {
    "golden_cases":
        (
            cell65_keyed_content_snapshot(
                cell65_active_nonexercise_golden,
                lambda row:
                    row[
                        "golden_case_id"
                    ],
            )
            ==
            cell65_keyed_content_snapshot(
                cell65_simulated_nonexercise_golden,
                lambda row:
                    row[
                        "golden_case_id"
                    ],
            )
        ),

    "auto_cases":
        (
            cell65_keyed_content_snapshot(
                cell65_active_nonexercise_auto,
                lambda row:
                    row[
                        "golden_case_id"
                    ],
            )
            ==
            cell65_keyed_content_snapshot(
                cell65_simulated_nonexercise_auto,
                lambda row:
                    row[
                        "golden_case_id"
                    ],
            )
        ),

    "groundability":
        (
            cell65_keyed_content_snapshot(
                cell65_active_nonexercise_groundability,
                lambda row:
                    row[
                        "golden_case_id"
                    ],
            )
            ==
            cell65_keyed_content_snapshot(
                cell65_simulated_nonexercise_groundability,
                lambda row:
                    row[
                        "golden_case_id"
                    ],
            )
        ),

    "task_design":
        (
            cell65_keyed_content_snapshot(
                cell65_active_nonexercise_design,
                lambda row:
                    row[
                        "golden_case_id"
                    ],
            )
            ==
            cell65_keyed_content_snapshot(
                cell65_simulated_nonexercise_design,
                lambda row:
                    row[
                        "golden_case_id"
                    ],
            )
        ),

    "task_drafts":
        (
            cell65_keyed_content_snapshot(
                cell65_active_nonexercise_tasks,
                lambda row:
                    row[
                        "task_draft_id"
                    ],
            )
            ==
            cell65_keyed_content_snapshot(
                cell65_simulated_nonexercise_tasks,
                lambda row:
                    row[
                        "task_draft_id"
                    ],
            )
        ),

    "targets":
        (
            cell65_keyed_content_snapshot(
                cell65_active_nonexercise_targets,
                lambda row:
                    row[
                        "target_evidence_id"
                    ],
            )
            ==
            cell65_keyed_content_snapshot(
                cell65_simulated_nonexercise_targets,
                lambda row:
                    row[
                        "target_evidence_id"
                    ],
            )
        ),

    "pilots":
        (
            cell65_keyed_content_snapshot(
                cell65_active_nonexercise_pilots,
                lambda row:
                    {
                        "source_type":
                            row[
                                "source_type"
                            ],

                        "stratum":
                            row[
                                "stratum"
                            ],
                    },
            )
            ==
            cell65_keyed_content_snapshot(
                cell65_simulated_nonexercise_pilots,
                lambda row:
                    {
                        "source_type":
                            row[
                                "source_type"
                            ],

                        "stratum":
                            row[
                                "stratum"
                            ],
                    },
            )
        ),
}


# ============================================================
# 22) V2 Exercise Slice semantic validation inside simulation
# ============================================================

cell65_v2_semantic_failures = []


for case_id in sorted(
    cell65_v2_case_ids
):

    target_row = (
        cell65_simulated_target_evidence_by_case_id.get(
            case_id
        )
    )


    if (
        target_row is None
    ):

        continue


    task_id = (
        target_row[
            "task_draft_id"
        ]
    )


    task_row = (
        cell65_simulated_task_draft_plans.get(
            task_id
        )
    )


    if (
        task_row is None
    ):

        continue


    if (
        task_row[
            "construction_eligibility"
        ]
        !=
        "READY_FOR_QUESTION_DRAFT"
    ):

        cell65_v2_semantic_failures.append(
            {
                "golden_case_id":
                    case_id,

                "reason":
                    "TASK_NOT_READY_FOR_QUESTION_DRAFT",
            }
        )


    if (
        task_row[
            "target_evidence_status"
        ]
        !=
        "SELECTED_AND_VALIDATED"
    ):

        cell65_v2_semantic_failures.append(
            {
                "golden_case_id":
                    case_id,

                "reason":
                    "TARGET_EVIDENCE_NOT_VALIDATED",
            }
        )


    if (
        target_row[
            "evidence_target_method"
        ]
        !=
        "USE_EXPLICIT_BOOK_EXERCISE_UNIT"
    ):

        cell65_v2_semantic_failures.append(
            {
                "golden_case_id":
                    case_id,

                "reason":
                    "V2_EVIDENCE_METHOD_MISMATCH",
            }
        )


    if not all(
        part[
            "role"
        ]
        ==
        "EXPLICIT_BOOK_EXERCISE_UNIT"

        for part
        in target_row[
            "target_parts"
        ]
    ):

        cell65_v2_semantic_failures.append(
            {
                "golden_case_id":
                    case_id,

                "reason":
                    "V2_PART_ROLE_MISMATCH",
            }
        )


# ============================================================
# 23) Bounded residual runtime dependency scan
#
# These references are NOT automatically errors.
#
# They must later be classified as:
#
#   CURRENT_STATE_REBUILD
#   DERIVED_REBUILD
#   HISTORICAL_PRESERVE
#   SAFE_NO_ACTION
# ============================================================

CELL65_CORE_ACTIVE_OBJECT_NAMES = {
    "golden_case_records",
    "golden_case_records_by_id",
    "auto_case_records",
    "case_groundability_precheck",
    "case_task_design",
    "task_draft_plan_rows",
    "task_draft_plans",
    "target_evidence_records",
    "target_evidence_by_case_id",
    "target_evidence_by_task_draft_id",
    "pilot_selection_rows",
}


CELL65_DEPENDENCY_NAME_TERMS = (
    "case",
    "task",
    "target",
    "pilot",
    "plan",
    "ready",
    "ground",
    "design",
    "candidate",
    "registry",
    "packet",
    "review",
    "question",
    "golden",
    "auto",
)


cell65_namespace_snapshot = list(
    globals().items()
)


cell65_residual_dependency_candidates = []


for variable_name, value in (
    cell65_namespace_snapshot
):

    lowered_name = (
        variable_name.lower()
    )


    if (
        variable_name.startswith(
            "cell"
        )
    ):

        continue


    if (
        variable_name.startswith(
            "staged_exercise_v2"
        )
    ):

        continue


    if (
        variable_name.startswith(
            "EXERCISE_V2"
        )
    ):

        continue


    if (
        variable_name
        in
        CELL65_CORE_ACTIVE_OBJECT_NAMES
    ):

        continue


    if not (
        any(
            term
            in
            lowered_name

            for term
            in CELL65_DEPENDENCY_NAME_TERMS
        )
    ):

        continue


    if not isinstance(
        value,
        (
            dict,
            list,
            tuple,
            set,
        ),
    ):

        continue


    try:

        top_level_length = len(
            value
        )

    except Exception:

        continue


    if (
        top_level_length
        >
        5000
    ):

        continue


    try:

        serialized_value = (
            stable_json_key(
                value
            )
        )

    except Exception:

        continue


    matched_case_ids = [
        legacy_id

        for legacy_id
        in cell65_legacy_exercise_case_ids

        if (
            legacy_id
            in
            serialized_value
        )
    ]


    matched_task_ids = [
        legacy_id

        for legacy_id
        in cell65_legacy_exercise_task_ids

        if (
            legacy_id
            in
            serialized_value
        )
    ]


    matched_target_ids = [
        legacy_id

        for legacy_id
        in cell65_legacy_exercise_target_ids

        if (
            legacy_id
            in
            serialized_value
        )
    ]


    if not (
        matched_case_ids
        or
        matched_task_ids
        or
        matched_target_ids
    ):

        continue


    cell65_residual_dependency_candidates.append(
        {
            "variable_name":
                variable_name,

            "type":
                type(
                    value
                ).__name__,

            "top_level_length":
                top_level_length,

            "matched_case_id_count":
                len(
                    matched_case_ids
                ),

            "matched_task_id_count":
                len(
                    matched_task_ids
                ),

            "matched_target_id_count":
                len(
                    matched_target_ids
                ),

            "classification":
                "REQUIRES_CURRENT_STATE_VS_HISTORICAL_CLASSIFICATION",
        }
    )


cell65_residual_dependency_candidates = sorted(
    cell65_residual_dependency_candidates,
    key=lambda row:
        row[
            "variable_name"
        ],
)


# ============================================================
# 24) Canonical indexes that must be rebuilt during any later
# real activation
# ============================================================

cell65_required_core_index_rebuilds = [
    "golden_case_records_by_id",
    "task_draft_plans",
    "target_evidence_by_case_id",
    "target_evidence_by_task_draft_id",
]


# ============================================================
# 25) Full simulated-state snapshot
# ============================================================

cell65_simulated_activation_payload = {
    "golden_case_records":
        cell65_simulated_golden_case_records,

    "auto_case_records":
        cell65_simulated_auto_case_records,

    "groundability":
        cell65_simulated_groundability_rows,

    "task_design":
        cell65_simulated_task_design_rows,

    "task_drafts":
        cell65_simulated_task_draft_plan_rows,

    "targets":
        cell65_simulated_target_evidence_records,

    "pilots":
        cell65_simulated_pilot_selection_rows,
}


cell65_simulated_activation_snapshot_sha256 = (
    cell65_sha(
        cell65_simulated_activation_payload
    )
)


# ============================================================
# 26) Structural simulation readiness
# ============================================================

cell65_structural_simulation_pass = (
    len(
        cell65_core_count_failures
    )
    ==
    0

    and

    len(
        cell65_duplicate_simulated_case_ids
    )
    ==
    0

    and

    len(
        cell65_duplicate_simulated_auto_case_ids
    )
    ==
    0

    and

    len(
        cell65_duplicate_simulated_groundability_case_ids
    )
    ==
    0

    and

    len(
        cell65_duplicate_simulated_design_case_ids
    )
    ==
    0

    and

    len(
        cell65_duplicate_simulated_task_ids
    )
    ==
    0

    and

    len(
        cell65_duplicate_simulated_target_ids
    )
    ==
    0

    and

    len(
        cell65_duplicate_simulated_pilot_strata
    )
    ==
    0

    and

    len(
        cell65_old_id_removal_failures
    )
    ==
    0

    and

    len(
        cell65_new_id_admission_failures
    )
    ==
    0

    and

    len(
        cell65_linkage_failures
    )
    ==
    0

    and

    len(
        cell65_pilot_linkage_failures
    )
    ==
    0

    and

    len(
        cell65_v2_semantic_failures
    )
    ==
    0

    and

    all(
        cell65_nonexercise_preservation.values()
    )
)


cell65_activation_readiness_status = (
    "CORE_SIMULATION_PASS_"
    "PENDING_RESIDUAL_DEPENDENCY_CLASSIFICATION"
    if (
        cell65_structural_simulation_pass
    )
    else
    "CORE_SIMULATION_FAILED"
)


# ============================================================
# 27) Freeze impact-audit contract
# ============================================================

EXERCISE_V2_ATOMIC_ACTIVATION_IMPACT_AUDIT_V1_DRAFT = {
    "version":
        "exercise_v2_atomic_activation_impact_audit_v1_draft",

    "status":
        cell65_activation_readiness_status,

    "legacy_slice": {
        "cases":
            len(
                cell65_legacy_exercise_cases
            ),

        "auto_cases":
            len(
                cell65_legacy_exercise_auto_cases
            ),

        "groundability":
            len(
                cell65_legacy_exercise_groundability
            ),

        "task_design":
            len(
                cell65_legacy_exercise_task_design
            ),

        "tasks":
            len(
                cell65_legacy_exercise_tasks
            ),

        "targets":
            len(
                cell65_legacy_exercise_targets
            ),

        "pilots":
            len(
                cell65_legacy_exercise_pilots
            ),
    },

    "v2_slice": {
        "cases":
            len(
                cell65_v2_cases
            ),

        "groundability":
            len(
                cell65_v2_groundability
            ),

        "task_design":
            len(
                cell65_v2_task_design
            ),

        "tasks":
            len(
                cell65_v2_tasks
            ),

        "targets":
            len(
                cell65_v2_targets
            ),

        "pilots":
            (
                1
                if cell65_v2_pilot is not None
                else 0
            ),
    },

    "simulated_counts":
        cell52_safe_clone(
            cell65_count_reconciliation
        ),

    "nonexercise_preservation":
        cell52_safe_clone(
            cell65_nonexercise_preservation
        ),

    "required_core_index_rebuilds":
        list(
            cell65_required_core_index_rebuilds
        ),

    "residual_dependency_candidate_count":
        len(
            cell65_residual_dependency_candidates
        ),

    "residual_dependency_status":
        (
            "CLASSIFICATION_REQUIRED"
            if (
                cell65_residual_dependency_candidates
            )
            else
            "NO_CANDIDATE_REFERENCES_FOUND_IN_BOUNDED_SCAN"
        ),

    "simulated_activation_snapshot_sha256":
        cell65_simulated_activation_snapshot_sha256,

    "actual_activation_status":
        "NOT_STARTED",

    "question_generation_status":
        "NOT_STARTED",

    "expected_answer_generation_status":
        "NOT_STARTED",

    "artifact_write_status":
        "NONE",
}


# ============================================================
# 28) No-mutation validation
# ============================================================

cell65_active_snapshots_after = {
    "golden_case_records":
        cell65_sha(
            golden_case_records
        ),

    "auto_case_records":
        cell65_sha(
            auto_case_records
        ),

    "case_groundability_precheck":
        cell65_sha(
            case_groundability_precheck
        ),

    "case_task_design":
        cell65_sha(
            case_task_design
        ),

    "task_draft_plan_rows":
        cell65_sha(
            task_draft_plan_rows
        ),

    "target_evidence_records":
        cell65_sha(
            target_evidence_records
        ),

    "pilot_selection_rows":
        cell65_sha(
            pilot_selection_rows
        ),
}


cell65_staging_snapshots_after = {
    "cases":
        cell65_sha(
            staged_exercise_v2_golden_case_records
        ),

    "groundability":
        cell65_sha(
            staged_exercise_v2_posttarget_groundability_precheck
        ),

    "task_design":
        cell65_sha(
            staged_exercise_v2_posttarget_case_task_design
        ),

    "tasks":
        cell65_sha(
            staged_exercise_v2_posttarget_task_draft_plan_rows
        ),

    "targets":
        cell65_sha(
            staged_exercise_v2_target_evidence_records
        ),

    "pilot":
        cell65_sha(
            staged_exercise_v2_pilot_replacement_row
        ),
}


# ============================================================
# 29) Gates
# ============================================================

cell65_gates = {
    "legacy_active_exercise_slice_is_exactly_38_cases":
        (
            len(
                cell65_legacy_exercise_cases
            )
            ==
            38
        ),

    "legacy_active_exercise_auto_slice_is_exactly_38":
        (
            len(
                cell65_legacy_exercise_auto_cases
            )
            ==
            38
        ),

    "legacy_active_exercise_groundability_slice_is_exactly_38":
        (
            len(
                cell65_legacy_exercise_groundability
            )
            ==
            38
        ),

    "legacy_active_exercise_task_design_slice_is_exactly_38":
        (
            len(
                cell65_legacy_exercise_task_design
            )
            ==
            38
        ),

    "legacy_active_exercise_task_slice_is_exactly_38":
        (
            len(
                cell65_legacy_exercise_tasks
            )
            ==
            38
        ),

    "legacy_active_exercise_target_slice_is_exactly_38":
        (
            len(
                cell65_legacy_exercise_targets
            )
            ==
            38
        ),

    "legacy_active_exercise_pilot_slice_is_exactly_1":
        (
            len(
                cell65_legacy_exercise_pilots
            )
            ==
            1
        ),

    "v2_replacement_slice_is_complete":
        (
            len(
                cell65_v2_cases
            )
            ==
            38

            and

            len(
                cell65_v2_groundability
            )
            ==
            38

            and

            len(
                cell65_v2_task_design
            )
            ==
            38

            and

            len(
                cell65_v2_tasks
            )
            ==
            38

            and

            len(
                cell65_v2_targets
            )
            ==
            38

            and

            cell65_v2_pilot
            is not None
        ),

    "legacy_and_v2_case_task_target_ids_are_disjoint":
        (
            len(
                cell65_case_id_collisions
            )
            ==
            0

            and

            len(
                cell65_task_id_collisions
            )
            ==
            0

            and

            len(
                cell65_target_id_collisions
            )
            ==
            0
        ),

    "all_simulated_core_counts_are_preserved":
        (
            len(
                cell65_core_count_failures
            )
            ==
            0
        ),

    "all_simulated_primary_keys_are_unique":
        (
            len(
                cell65_duplicate_simulated_case_ids
            )
            ==
            0

            and

            len(
                cell65_duplicate_simulated_auto_case_ids
            )
            ==
            0

            and

            len(
                cell65_duplicate_simulated_groundability_case_ids
            )
            ==
            0

            and

            len(
                cell65_duplicate_simulated_design_case_ids
            )
            ==
            0

            and

            len(
                cell65_duplicate_simulated_task_ids
            )
            ==
            0

            and

            len(
                cell65_duplicate_simulated_target_ids
            )
            ==
            0

            and

            len(
                cell65_duplicate_simulated_pilot_strata
            )
            ==
            0
        ),

    "all_legacy_ids_are_absent_from_simulated_current_state":
        (
            len(
                cell65_old_id_removal_failures
            )
            ==
            0
        ),

    "all_v2_ids_are_admitted_to_every_required_simulated_registry":
        (
            len(
                cell65_new_id_admission_failures
            )
            ==
            0
        ),

    "all_simulated_case_task_target_links_are_valid":
        (
            len(
                cell65_linkage_failures
            )
            ==
            0
        ),

    "simulated_pilot_registry_contains_exactly_one_valid_exercise_pilot":
        (
            len(
                cell65_simulated_exercise_pilots
            )
            ==
            1

            and

            len(
                cell65_pilot_linkage_failures
            )
            ==
            0
        ),

    "all_nonexercise_content_is_preserved_exactly_by_identity":
        all(
            cell65_nonexercise_preservation.values()
        ),

    "all_v2_exercise_semantics_survive_simulated_activation":
        (
            len(
                cell65_v2_semantic_failures
            )
            ==
            0
        ),

    "bounded_residual_dependency_scan_completed":
        isinstance(
            cell65_residual_dependency_candidates,
            list,
        ),

    "core_simulation_passes_but_activation_waits_for_dependency_classification":
        (
            cell65_activation_readiness_status
            ==
            "CORE_SIMULATION_PASS_"
            "PENDING_RESIDUAL_DEPENDENCY_CLASSIFICATION"
        ),

    "all_active_core_registries_remain_byte_stable":
        (
            cell65_active_snapshots_before
            ==
            cell65_active_snapshots_after
        ),

    "all_existing_v2_staging_inputs_remain_byte_stable":
        (
            cell65_staging_snapshots_before
            ==
            cell65_staging_snapshots_after
        ),

    "actual_activation_not_started":
        (
            EXERCISE_V2_ATOMIC_ACTIVATION_IMPACT_AUDIT_V1_DRAFT[
                "actual_activation_status"
            ]
            ==
            "NOT_STARTED"
        ),

    "question_generation_not_started":
        (
            EXERCISE_V2_ATOMIC_ACTIVATION_IMPACT_AUDIT_V1_DRAFT[
                "question_generation_status"
            ]
            ==
            "NOT_STARTED"
        ),

    "expected_answer_generation_not_started":
        (
            EXERCISE_V2_ATOMIC_ACTIVATION_IMPACT_AUDIT_V1_DRAFT[
                "expected_answer_generation_status"
            ]
            ==
            "NOT_STARTED"
        ),

    "no_artifact_write":
        (
            EXERCISE_V2_ATOMIC_ACTIVATION_IMPACT_AUDIT_V1_DRAFT[
                "artifact_write_status"
            ]
            ==
            "NONE"
        ),
}


cell_65_pass = all(
    cell65_gates.values()
)


# ============================================================
# 30) Output
# ============================================================

print(
    "=== EXERCISE V2 ATOMIC ACTIVATION SIMULATION — REPAIRED ==="
)


print(
    "\n=== LEGACY ACTIVE SLICE ==="
)


print(
    "Cases:",
    len(
        cell65_legacy_exercise_cases
    )
)


print(
    "Auto:",
    len(
        cell65_legacy_exercise_auto_cases
    )
)


print(
    "Groundability:",
    len(
        cell65_legacy_exercise_groundability
    )
)


print(
    "Task Design:",
    len(
        cell65_legacy_exercise_task_design
    )
)


print(
    "Tasks:",
    len(
        cell65_legacy_exercise_tasks
    )
)


print(
    "Targets:",
    len(
        cell65_legacy_exercise_targets
    )
)


print(
    "Pilots:",
    len(
        cell65_legacy_exercise_pilots
    )
)


print(
    "\n=== V2 REPLACEMENT SLICE ==="
)


print(
    "Cases:",
    len(
        cell65_v2_cases
    )
)


print(
    "Groundability:",
    len(
        cell65_v2_groundability
    )
)


print(
    "Task Design:",
    len(
        cell65_v2_task_design
    )
)


print(
    "Tasks:",
    len(
        cell65_v2_tasks
    )
)


print(
    "Targets:",
    len(
        cell65_v2_targets
    )
)


print(
    "Pilots:",
    1
    if cell65_v2_pilot is not None
    else 0
)


print(
    "\n=== COUNT RECONCILIATION ==="
)


for registry_name, counts in (
    cell65_count_reconciliation.items()
):

    print(
        "-",
        registry_name,
        ":",
        counts,
    )


print(
    "\nCount failures:",
    cell65_core_count_failures
)


print(
    "\n=== PRIMARY KEY / ID AUDIT ==="
)


print(
    "Case ID collisions:",
    len(
        cell65_case_id_collisions
    )
)


print(
    "Task ID collisions:",
    len(
        cell65_task_id_collisions
    )
)


print(
    "Target ID collisions:",
    len(
        cell65_target_id_collisions
    )
)


print(
    "Old-ID removal failures:",
    len(
        cell65_old_id_removal_failures
    )
)


print(
    "New-ID admission failures:",
    len(
        cell65_new_id_admission_failures
    )
)


print(
    "Linkage failures:",
    len(
        cell65_linkage_failures
    )
)


print(
    "Pilot linkage failures:",
    len(
        cell65_pilot_linkage_failures
    )
)


print(
    "V2 semantic failures:",
    len(
        cell65_v2_semantic_failures
    )
)


print(
    "\n=== NON-EXERCISE PRESERVATION — IDENTITY KEYED ==="
)


for family_name, preserved in (
    cell65_nonexercise_preservation.items()
):

    print(
        "-",
        family_name,
        ":",
        "PASS"
        if preserved
        else "FAIL"
    )


print(
    "\n=== REQUIRED CORE INDEX REBUILDS ==="
)


for name in (
    cell65_required_core_index_rebuilds
):

    print(
        "-",
        name
    )


print(
    "\n=== RESIDUAL LEGACY DEPENDENCY CANDIDATES ==="
)


print(
    "Candidate count:",
    len(
        cell65_residual_dependency_candidates
    )
)


for row in (
    cell65_residual_dependency_candidates
):

    print(
        "-",
        row
    )


print(
    "\n=== ACTIVATION READINESS ==="
)


print(
    "Structural simulation pass:",
    cell65_structural_simulation_pass
)


print(
    "Status:",
    cell65_activation_readiness_status
)


print(
    "Simulated Activation Snapshot SHA256:",
    cell65_simulated_activation_snapshot_sha256
)


print(
    "\n=== ACTIVE COUNTS — STILL UNCHANGED ==="
)


print(
    "Golden:",
    len(
        golden_case_records
    )
)


print(
    "Auto:",
    len(
        cell65_active_auto_rows
    )
)


print(
    "Groundability:",
    len(
        cell65_active_groundability_rows
    )
)


print(
    "Task Design:",
    len(
        cell65_active_task_design_rows
    )
)


print(
    "Tasks:",
    len(
        task_draft_plan_rows
    )
)


print(
    "Targets:",
    len(
        target_evidence_records
    )
)


print(
    "Pilots:",
    len(
        pilot_selection_rows
    )
)


print(
    "\n=== CODE CELL 65 GATES ==="
)


for gate_name, passed in (
    cell65_gates.items()
):

    print(
        gate_name,
        ":",
        "PASS"
        if passed
        else "FAIL"
    )


print(
    "\nFAILED GATES:",
    [
        gate_name

        for gate_name, passed
        in cell65_gates.items()

        if not passed
    ]
)


print(
    "\nCODE CELL 65 GATE:",
    "PASS"
    if cell_65_pass
    else "FAIL"
)


assert cell_65_pass, (
    "Atomic Exercise V2 activation simulation remains "
    "unresolved. Do not mutate active registries. Use the "
    "explicit failed gates and diagnostic sections from this "
    "SAME Code Cell 65 before activation."
)
```

<h3>Residual Legacy Dependency Classification and Activation Blocker Resolution</h3>

<p>
Cell 65 ثابت کرد که جایگزینی کامل Exercise Slice از نظر ساختاری قابل انجام است:
تعداد Registryها حفظ می‌شود، ID collision وجود ندارد، Indexها قابل بازسازی‌اند،
تمام Non-Exercise content بدون تغییر باقی می‌ماند و Pilot جدید نیز در Simulated State معتبر است.
</p>

<p>
تنها Blocker باقی‌مانده 26 Runtime Object است که هنوز به Legacy Exercise Case/Task/Target IDs
اشاره دارند.
وجود این Referenceها به‌تنهایی Error نیست.
</p>

<p>
هر Object باید بر اساس Source واقعی تعریف و مصرفش در Notebook در یکی از چهار گروه قرار گیرد:
</p>

<ul>
<li><code>DERIVED_REBUILD</code> — مشتق مستقیم از Current Active State و هنگام Activation باید بازسازی شود؛</li>
<li><code>CURRENT_STATE_REPLACE</code> — خودش بخشی از Current State عملیاتی است و باید با نسخه‌ی V2 جایگزین شود؛</li>
<li><code>HISTORICAL_PRESERVE</code> — Evidence/Audit/Lineage مربوط به Legacy repair است و باید عمداً حفظ شود؛</li>
<li><code>SAFE_NO_ACTION</code> — Scratch/loop/local diagnostic object است و در Canonical State نقشی ندارد.</li>
</ul>

<p>
این Cell ابتدا محل تعریف، Source snippet و مصرف بعدی هر یک از 26 Object را استخراج می‌کند.
سپس Classification فقط وقتی پذیرفته می‌شود که با Source evidence سازگار باشد و هیچ Object
در وضعیت <code>UNRESOLVED</code> باقی نماند.
</p>

<p>
هنوز هیچ Active Registry تغییر نمی‌کند.
</p>

</div>

```python
# Code Cell 66 — RUN-ALL STABLE REPAIR

# ============================================================
# Residual Dependency Classification
#
# Repair:
#   - Cell65 runtime universe is authoritative.
#   - Do NOT hard-code 27 residual objects.
#   - indexed_target is classified only if Cell65 actually
#     includes it.
#   - Use the corrected dependency policy:
#
#       DERIVED_REBUILD      = 5
#       HISTORICAL_PRESERVE  = 13 possible
#       SAFE_NO_ACTION       = 9 possible
#
#     Current 26-object universe therefore resolves to:
#
#       DERIVED_REBUILD      = 5
#       HISTORICAL_PRESERVE  = 13
#       SAFE_NO_ACTION       = 8
#
# READ ONLY.
# NO activation.
# NO derived rebuild execution.
# NO question generation.
# NO artifact write.
# ============================================================


# ============================================================
# 1) Helper
# ============================================================

def cell66_sha(value):

    return hashlib.sha256(
        stable_json_key(
            value
        ).encode(
            "utf-8"
        )
    ).hexdigest()


# ============================================================
# 2) Snapshot active + validated staging state
# ============================================================

cell66_active_snapshot_before = {
    "golden":
        cell66_sha(golden_case_records),

    "auto":
        cell66_sha(auto_case_records),

    "groundability":
        cell66_sha(case_groundability_precheck),

    "task_design":
        cell66_sha(case_task_design),

    "tasks":
        cell66_sha(task_draft_plan_rows),

    "targets":
        cell66_sha(target_evidence_records),

    "pilots":
        cell66_sha(pilot_selection_rows),
}


cell66_staging_snapshot_before = {
    "cases":
        cell66_sha(
            staged_exercise_v2_golden_case_records
        ),

    "groundability":
        cell66_sha(
            staged_exercise_v2_posttarget_groundability_precheck
        ),

    "task_design":
        cell66_sha(
            staged_exercise_v2_posttarget_case_task_design
        ),

    "tasks":
        cell66_sha(
            staged_exercise_v2_posttarget_task_draft_plan_rows
        ),

    "targets":
        cell66_sha(
            staged_exercise_v2_target_evidence_records
        ),

    "pilot":
        cell66_sha(
            staged_exercise_v2_pilot_replacement_row
        ),
}


# ============================================================
# 3) AUTHORITATIVE residual universe = exact Cell65 output
# ============================================================

cell66_residual_candidates = (
    cell52_safe_clone(
        cell65_residual_dependency_candidates
    )
)


cell66_candidate_by_name = {
    row["variable_name"]:
        row

    for row
    in cell66_residual_candidates
}


cell66_residual_names = sorted(
    cell66_candidate_by_name
)


cell66_residual_name_set = set(
    cell66_residual_names
)


cell66_runtime_missing_objects = sorted(
    name

    for name
    in cell66_residual_names

    if name not in globals()
)


# ============================================================
# 4) Corrected dependency policy
#
# IMPORTANT:
# These are all KNOWN possible residual names.
# The active classification is filtered to the exact Cell65
# runtime universe.
# ============================================================

CELL66_MASTER_CLASSIFICATION_POLICY = {

    # --------------------------------------------------------
    # TRUE CURRENT-STATE DERIVED OBJECTS — exactly five
    # --------------------------------------------------------

    "candidate_hash_to_case_ids": {
        "classification":
            "DERIVED_REBUILD",

        "rationale":
            "DERIVED_INDEX_OVER_ACTIVE_GOLDEN_CASES",
    },

    "cross_stratum_repeated_candidate_hashes": {
        "classification":
            "DERIVED_REBUILD",

        "rationale":
            "DERIVED_CROSS_STRATUM_CASE_HASH_ANALYSIS",
    },

    "question_drafting_pilot_by_id": {
        "classification":
            "DERIVED_REBUILD",

        "rationale":
            "CURRENT_QUESTION_PILOT_INDEX_REFERENCES_ACTIVE_CASE_TASK_TARGET",
    },

    "question_drafting_pilot_packets": {
        "classification":
            "DERIVED_REBUILD",

        "rationale":
            "CURRENT_QUESTION_PILOT_PACKETS_REFERENCE_ACTIVE_CASE_TASK_TARGET",
    },

    "target_id_counts": {
        "classification":
            "DERIVED_REBUILD",

        "rationale":
            "DERIVED_COUNT_OVER_ACTIVE_TARGET_IDS",
    },


    # --------------------------------------------------------
    # HISTORICAL / AUDIT / PRE-REPAIR EVIDENCE
    # --------------------------------------------------------

    "exercise_selected_cases": {
        "classification":
            "HISTORICAL_PRESERVE",
        "rationale":
            "LEGACY_EXERCISE_SELECTION_EVIDENCE",
    },

    "failed_exercise_pilot": {
        "classification":
            "HISTORICAL_PRESERVE",
        "rationale":
            "FAILED_LEGACY_PILOT_AUDIT_EVIDENCE",
    },

    "legacy_case_ids": {
        "classification":
            "HISTORICAL_PRESERVE",
        "rationale":
            "LEGACY_CASE_LINEAGE",
    },

    "legacy_case_mapping_counts": {
        "classification":
            "HISTORICAL_PRESERVE",
        "rationale":
            "LEGACY_TO_V2_MAPPING_AUDIT",
    },

    "legacy_cases_inside_core": {
        "classification":
            "HISTORICAL_PRESERVE",
        "rationale":
            "LEGACY_CORE_MEMBERSHIP_AUDIT",
    },

    "legacy_to_proposed_exercise_case_mapping": {
        "classification":
            "HISTORICAL_PRESERVE",
        "rationale":
            "LEGACY_TO_STRUCTURAL_UNIT_LINEAGE",
    },

    "pilot_full_inspection_rows": {
        "classification":
            "HISTORICAL_PRESERVE",
        "rationale":
            "PRE_REPAIR_PILOT_INSPECTION",
    },

    "pilot_semantic_review_rows": {
        "classification":
            "HISTORICAL_PRESERVE",
        "rationale":
            "PRE_REPAIR_SEMANTIC_REVIEW",
    },

    "proposed_exercise_unit_cases": {
        "classification":
            "HISTORICAL_PRESERVE",
        "rationale":
            "INTERMEDIATE_STRUCTURAL_REPAIR_EVIDENCE",
    },

    "proposed_exercise_unit_cases_by_id": {
        "classification":
            "HISTORICAL_PRESERVE",
        "rationale":
            "INDEX_OF_INTERMEDIATE_STRUCTURAL_REPAIR_EVIDENCE",
    },

    # Corrected after the later dependency audit:
    "pilot_rows_by_token_size": {
        "classification":
            "HISTORICAL_PRESERVE",
        "rationale":
            "HISTORICAL_PILOT_SELECTION_DIAGNOSTIC_VIEW",
    },

    "ready_case_plans": {
        "classification":
            "HISTORICAL_PRESERVE",
        "rationale":
            "HISTORICAL_READY_PLAN_SNAPSHOT_NOT_LIVE_DEPENDENCY",
    },

    "target_method_readiness": {
        "classification":
            "HISTORICAL_PRESERVE",
        "rationale":
            "HISTORICAL_TARGET_METHOD_READINESS_DIAGNOSTIC",
    },


    # --------------------------------------------------------
    # SAFE transient / scratch values
    # --------------------------------------------------------

    "case": {
        "classification":
            "SAFE_NO_ACTION",
        "rationale":
            "TRANSIENT_LOOP_VALUE",
    },

    "case_record": {
        "classification":
            "SAFE_NO_ACTION",
        "rationale":
            "TRANSIENT_CASE_VALUE",
    },

    "groundability_record": {
        "classification":
            "SAFE_NO_ACTION",
        "rationale":
            "TRANSIENT_GROUNDABILITY_VALUE",
    },

    "indexed_target": {
        "classification":
            "SAFE_NO_ACTION",
        "rationale":
            "TRANSIENT_TARGET_LOOKUP_VALUE",
    },

    "matching_targets": {
        "classification":
            "SAFE_NO_ACTION",
        "rationale":
            "TRANSIENT_FILTER_RESULT",
    },

    "proposed_case": {
        "classification":
            "SAFE_NO_ACTION",
        "rationale":
            "TRANSIENT_PROPOSED_CASE_VALUE",
    },

    "ready_row": {
        "classification":
            "SAFE_NO_ACTION",
        "rationale":
            "TRANSIENT_READY_ROW",
    },

    "registry_row": {
        "classification":
            "SAFE_NO_ACTION",
        "rationale":
            "TRANSIENT_REGISTRY_ROW",
    },

    "task_design_record": {
        "classification":
            "SAFE_NO_ACTION",
        "rationale":
            "TRANSIENT_TASK_DESIGN_VALUE",
    },
}


# ============================================================
# 5) Filter master policy to EXACT Cell65 universe
# ============================================================

CELL66_EXPLICIT_CLASSIFICATION = {
    name:
        cell52_safe_clone(
            CELL66_MASTER_CLASSIFICATION_POLICY[name]
        )

    for name
    in cell66_residual_names

    if name
    in CELL66_MASTER_CLASSIFICATION_POLICY
}


cell66_classification_name_set = set(
    CELL66_EXPLICIT_CLASSIFICATION
)


cell66_missing_classification_names = sorted(
    cell66_residual_name_set
    -
    cell66_classification_name_set
)


# Because the policy is filtered to Cell65, extras must always
# remain empty.
cell66_extra_classification_names = sorted(
    cell66_classification_name_set
    -
    cell66_residual_name_set
)


# ============================================================
# 6) Allowed vocabulary
# ============================================================

CELL66_ALLOWED_CLASSIFICATIONS = {
    "DERIVED_REBUILD",
    "HISTORICAL_PRESERVE",
    "SAFE_NO_ACTION",
    "CURRENT_STATE_REPLACE",
}


cell66_invalid_classification_values = [
    {
        "variable_name":
            name,

        "classification":
            policy.get(
                "classification"
            ),
    }

    for name, policy
    in CELL66_EXPLICIT_CLASSIFICATION.items()

    if (
        policy.get(
            "classification"
        )
        not in
        CELL66_ALLOWED_CLASSIFICATIONS
    )
]


# ============================================================
# 7) Materialize classification rows
# ============================================================

cell66_classification_rows = []


for name in cell66_residual_names:

    candidate = (
        cell66_candidate_by_name[
            name
        ]
    )

    policy = (
        CELL66_EXPLICIT_CLASSIFICATION.get(
            name
        )
    )


    cell66_classification_rows.append(
        {
            "variable_name":
                name,

            "classification":
                (
                    policy[
                        "classification"
                    ]
                    if policy
                    else
                    "UNRESOLVED"
                ),

            "rationale":
                (
                    policy[
                        "rationale"
                    ]
                    if policy
                    else
                    "NO_POLICY"
                ),

            "runtime_exists":
                name in globals(),

            "type":
                candidate.get(
                    "type"
                ),

            "top_level_length":
                candidate.get(
                    "top_level_length"
                ),

            "matched_case_id_count":
                candidate.get(
                    "matched_case_id_count",
                    0,
                ),

            "matched_task_id_count":
                candidate.get(
                    "matched_task_id_count",
                    0,
                ),

            "matched_target_id_count":
                candidate.get(
                    "matched_target_id_count",
                    0,
                ),
        }
    )


cell66_unresolved_classifications = [
    row

    for row
    in cell66_classification_rows

    if (
        row[
            "classification"
        ]
        ==
        "UNRESOLVED"
    )
]


# ============================================================
# 8) Classification families
# ============================================================

cell66_classification_counts = Counter(
    row[
        "classification"
    ]

    for row
    in cell66_classification_rows
)


cell66_derived_rebuild_objects = sorted(
    row["variable_name"]

    for row
    in cell66_classification_rows

    if (
        row["classification"]
        ==
        "DERIVED_REBUILD"
    )
)


cell66_historical_preserve_objects = sorted(
    row["variable_name"]

    for row
    in cell66_classification_rows

    if (
        row["classification"]
        ==
        "HISTORICAL_PRESERVE"
    )
)


cell66_safe_no_action_objects = sorted(
    row["variable_name"]

    for row
    in cell66_classification_rows

    if (
        row["classification"]
        ==
        "SAFE_NO_ACTION"
    )
)


cell66_current_state_replace_objects = sorted(
    row["variable_name"]

    for row
    in cell66_classification_rows

    if (
        row["classification"]
        ==
        "CURRENT_STATE_REPLACE"
    )
)


# ============================================================
# 9) Exact TRUE derived rebuild contract
# ============================================================

CELL66_REQUIRED_DERIVED_REBUILDS = {
    "candidate_hash_to_case_ids",
    "cross_stratum_repeated_candidate_hashes",
    "question_drafting_pilot_by_id",
    "question_drafting_pilot_packets",
    "target_id_counts",
}


# Current-state objects that must be rebuilt before any
# question text can be generated. Historical diagnostics are
# intentionally excluded from this set.
CELL66_REQUIRED_PREQUESTION_REBUILDS = {
    "question_drafting_pilot_by_id",
    "question_drafting_pilot_packets",
}


cell66_missing_derived_rebuilds = sorted(
    CELL66_REQUIRED_DERIVED_REBUILDS
    -
    set(
        cell66_derived_rebuild_objects
    )
)


cell66_unexpected_derived_rebuilds = sorted(
    set(
        cell66_derived_rebuild_objects
    )
    -
    CELL66_REQUIRED_DERIVED_REBUILDS
)


# ============================================================
# 10) Historical corrections explicitly enforced
# ============================================================

CELL66_CORRECTED_HISTORICAL_OBJECTS = {
    "pilot_rows_by_token_size",
    "ready_case_plans",
    "target_method_readiness",
}


cell66_missing_corrected_historical_objects = sorted(
    (
        CELL66_CORRECTED_HISTORICAL_OBJECTS
        &
        cell66_residual_name_set
    )
    -
    set(
        cell66_historical_preserve_objects
    )
)


# ============================================================
# 11) Canonical indexes still rebuilt atomically
# ============================================================

cell66_canonical_index_rebuilds = sorted(
    cell65_required_core_index_rebuilds
)


CELL66_REQUIRED_CANONICAL_INDEX_REBUILDS = {
    "golden_case_records_by_id",
    "task_draft_plans",
    "target_evidence_by_case_id",
    "target_evidence_by_task_draft_id",
}


cell66_missing_canonical_index_rebuilds = sorted(
    CELL66_REQUIRED_CANONICAL_INDEX_REBUILDS
    -
    set(
        cell66_canonical_index_rebuilds
    )
)


cell66_unexpected_canonical_index_rebuilds = sorted(
    set(
        cell66_canonical_index_rebuilds
    )
    -
    CELL66_REQUIRED_CANONICAL_INDEX_REBUILDS
)


# ============================================================
# 12) Canonical collision protection
# ============================================================

CELL66_CANONICAL_CURRENT_STATE_NAMES = {
    "golden_case_records",
    "golden_case_records_by_id",
    "auto_case_records",
    "case_groundability_precheck",
    "case_task_design",
    "task_draft_plan_rows",
    "task_draft_plans",
    "target_evidence_records",
    "target_evidence_by_case_id",
    "target_evidence_by_task_draft_id",
    "pilot_selection_rows",
}


cell66_historical_core_conflicts = sorted(
    set(
        cell66_historical_preserve_objects
    )
    &
    CELL66_CANONICAL_CURRENT_STATE_NAMES
)


cell66_safe_no_action_core_conflicts = sorted(
    set(
        cell66_safe_no_action_objects
    )
    &
    CELL66_CANONICAL_CURRENT_STATE_NAMES
)


# ============================================================
# 13) Historical lineage must actually preserve legacy refs
# ============================================================

cell66_historical_reference_total = sum(
    (
        row.get(
            "matched_case_id_count",
            0,
        )
        +
        row.get(
            "matched_task_id_count",
            0,
        )
        +
        row.get(
            "matched_target_id_count",
            0,
        )
    )

    for row
    in cell66_classification_rows

    if (
        row[
            "classification"
        ]
        ==
        "HISTORICAL_PRESERVE"
    )
)


cell66_historical_preserve_intentionally_contains_legacy_refs = (
    cell66_historical_reference_total
    >
    0
)


# ============================================================
# 14) No rebuild / safe conflict
# ============================================================

cell66_safe_rebuild_conflicts = sorted(
    set(
        cell66_safe_no_action_objects
    )
    &
    (
        CELL66_REQUIRED_DERIVED_REBUILDS
        |
        CELL66_REQUIRED_CANONICAL_INDEX_REBUILDS
    )
)


# ============================================================
# 15) Optional source provenance diagnostic
# ============================================================

def cell66_optional_source_occurrences(
    variable_name,
):

    occurrences = []


    for code_cell in cell49_code_cells:

        source = (
            code_cell.get(
                "source",
                ""
            )
        )


        if variable_name in source:

            occurrences.append(
                code_cell.get(
                    "cell_index"
                )
            )


    return occurrences


cell66_optional_source_evidence = {
    name:
        cell66_optional_source_occurrences(
            name
        )

    for name
    in cell66_residual_names
}


cell66_objects_with_source_occurrence = sorted(
    name

    for name, hits
    in cell66_optional_source_evidence.items()

    if hits
)


cell66_objects_without_source_occurrence = sorted(
    name

    for name, hits
    in cell66_optional_source_evidence.items()

    if not hits
)


# ============================================================
# 16) Activation rebuild plan
# ============================================================

cell66_activation_rebuild_plan = {
    "canonical_index_rebuilds":
        list(
            cell66_canonical_index_rebuilds
        ),

    "derived_runtime_rebuilds":
        list(
            cell66_derived_rebuild_objects
        ),

    "current_state_replacements":
        list(
            cell66_current_state_replace_objects
        ),

    "historical_preserve":
        list(
            cell66_historical_preserve_objects
        ),

    "safe_no_action":
        list(
            cell66_safe_no_action_objects
        ),
}


# ============================================================
# 17) Classification completeness
#
# KEY REPAIR:
# No hard-coded "27".
# Exact Cell65 runtime universe is authoritative.
# ============================================================

cell66_dependency_classification_complete = (
    len(
        cell66_residual_candidates
    )
    ==
    len(
        cell66_residual_name_set
    )

    and

    len(
        cell66_residual_name_set
    )
    >
    0

    and

    len(
        cell66_missing_classification_names
    )
    ==
    0

    and

    len(
        cell66_extra_classification_names
    )
    ==
    0

    and

    len(
        cell66_runtime_missing_objects
    )
    ==
    0

    and

    len(
        cell66_invalid_classification_values
    )
    ==
    0

    and

    len(
        cell66_unresolved_classifications
    )
    ==
    0

    and

    len(
        cell66_missing_derived_rebuilds
    )
    ==
    0

    and

    len(
        cell66_unexpected_derived_rebuilds
    )
    ==
    0

    and

    len(
        cell66_missing_corrected_historical_objects
    )
    ==
    0

    and

    len(
        cell66_missing_canonical_index_rebuilds
    )
    ==
    0

    and

    len(
        cell66_unexpected_canonical_index_rebuilds
    )
    ==
    0

    and

    len(
        cell66_historical_core_conflicts
    )
    ==
    0

    and

    len(
        cell66_safe_no_action_core_conflicts
    )
    ==
    0

    and

    len(
        cell66_safe_rebuild_conflicts
    )
    ==
    0

    and

    cell66_historical_preserve_intentionally_contains_legacy_refs
)


cell66_dependency_classification_status = (
    "ALL_RESIDUAL_DEPENDENCIES_CLASSIFIED"

    if (
        cell66_dependency_classification_complete
    )

    else
    "DEPENDENCY_CLASSIFICATION_INCOMPLETE"
)


# ============================================================
# 18) Freeze contract
# ============================================================

EXERCISE_V2_RESIDUAL_DEPENDENCY_CLASSIFICATION_V1_DRAFT = {
    "version":
        "exercise_v2_residual_dependency_classification_v1_draft",

    "status":
        cell66_dependency_classification_status,

    "candidate_count":
        len(
            cell66_residual_candidates
        ),

    "classification_counts":
        dict(
            cell66_classification_counts
        ),

    "canonical_index_rebuilds":
        list(
            cell66_canonical_index_rebuilds
        ),

    "derived_runtime_rebuilds":
        list(
            cell66_derived_rebuild_objects
        ),

    "current_state_replacements":
        list(
            cell66_current_state_replace_objects
        ),

    "historical_preserve":
        list(
            cell66_historical_preserve_objects
        ),

    "safe_no_action":
        list(
            cell66_safe_no_action_objects
        ),

    "actual_activation_status":
        "NOT_STARTED",

    "derived_rebuild_execution_status":
        "NOT_STARTED",

    "question_generation_status":
        "NOT_STARTED",

    "expected_answer_generation_status":
        "NOT_STARTED",

    "artifact_write_status":
        "NONE",
}


# ============================================================
# 19) No-mutation recheck
# ============================================================

cell66_active_snapshot_after = {
    "golden":
        cell66_sha(golden_case_records),

    "auto":
        cell66_sha(auto_case_records),

    "groundability":
        cell66_sha(case_groundability_precheck),

    "task_design":
        cell66_sha(case_task_design),

    "tasks":
        cell66_sha(task_draft_plan_rows),

    "targets":
        cell66_sha(target_evidence_records),

    "pilots":
        cell66_sha(pilot_selection_rows),
}


cell66_staging_snapshot_after = {
    "cases":
        cell66_sha(
            staged_exercise_v2_golden_case_records
        ),

    "groundability":
        cell66_sha(
            staged_exercise_v2_posttarget_groundability_precheck
        ),

    "task_design":
        cell66_sha(
            staged_exercise_v2_posttarget_case_task_design
        ),

    "tasks":
        cell66_sha(
            staged_exercise_v2_posttarget_task_draft_plan_rows
        ),

    "targets":
        cell66_sha(
            staged_exercise_v2_target_evidence_records
        ),

    "pilot":
        cell66_sha(
            staged_exercise_v2_pilot_replacement_row
        ),
}


# ============================================================
# 20) Gates
# ============================================================

cell66_gates = {

    "cell65_residual_universe_is_nonempty_and_unique":
        (
            len(
                cell66_residual_candidates
            )
            ==
            len(
                cell66_residual_name_set
            )

            and

            len(
                cell66_residual_name_set
            )
            >
            0
        ),

    "classification_registry_covers_exactly_current_cell65_universe":
        (
            not cell66_missing_classification_names
            and
            not cell66_extra_classification_names
        ),

    "all_residual_runtime_objects_exist":
        (
            not cell66_runtime_missing_objects
        ),

    "all_classification_values_are_valid":
        (
            not cell66_invalid_classification_values
        ),

    "no_dependency_is_unresolved":
        (
            not cell66_unresolved_classifications
        ),

    "exact_five_true_derived_dependencies_are_rebuilt":
        (
            not cell66_missing_derived_rebuilds
            and
            not cell66_unexpected_derived_rebuilds
            and
            len(
                cell66_derived_rebuild_objects
            )
            ==
            5
        ),

    "three_previously_overclassified_objects_are_historical":
        (
            not cell66_missing_corrected_historical_objects
        ),

    "canonical_index_rebuild_contract_is_exact":
        (
            not cell66_missing_canonical_index_rebuilds
            and
            not cell66_unexpected_canonical_index_rebuilds
        ),

    "historical_and_scratch_objects_do_not_overlap_canonical_state":
        (
            not cell66_historical_core_conflicts
            and
            not cell66_safe_no_action_core_conflicts
        ),

    "historical_preserve_retains_real_legacy_lineage":
        (
            cell66_historical_preserve_intentionally_contains_legacy_refs
        ),

    "safe_objects_are_not_rebuild_dependencies":
        (
            not cell66_safe_rebuild_conflicts
        ),

    "dependency_classification_is_complete":
        (
            cell66_dependency_classification_status
            ==
            "ALL_RESIDUAL_DEPENDENCIES_CLASSIFIED"
        ),

    "cell65_structural_simulation_remains_pass":
        (
            cell65_structural_simulation_pass
            is True

            and

            cell65_activation_readiness_status
            ==
            "CORE_SIMULATION_PASS_"
            "PENDING_RESIDUAL_DEPENDENCY_CLASSIFICATION"
        ),

    "active_state_remains_byte_stable":
        (
            cell66_active_snapshot_before
            ==
            cell66_active_snapshot_after
        ),

    "validated_v2_staging_remains_byte_stable":
        (
            cell66_staging_snapshot_before
            ==
            cell66_staging_snapshot_after
        ),

    "activation_not_started":
        (
            EXERCISE_V2_RESIDUAL_DEPENDENCY_CLASSIFICATION_V1_DRAFT[
                "actual_activation_status"
            ]
            ==
            "NOT_STARTED"
        ),

    "derived_rebuild_execution_not_started":
        (
            EXERCISE_V2_RESIDUAL_DEPENDENCY_CLASSIFICATION_V1_DRAFT[
                "derived_rebuild_execution_status"
            ]
            ==
            "NOT_STARTED"
        ),

    "question_generation_not_started":
        (
            EXERCISE_V2_RESIDUAL_DEPENDENCY_CLASSIFICATION_V1_DRAFT[
                "question_generation_status"
            ]
            ==
            "NOT_STARTED"
        ),

    "expected_answer_generation_not_started":
        (
            EXERCISE_V2_RESIDUAL_DEPENDENCY_CLASSIFICATION_V1_DRAFT[
                "expected_answer_generation_status"
            ]
            ==
            "NOT_STARTED"
        ),

    "no_artifact_write":
        (
            EXERCISE_V2_RESIDUAL_DEPENDENCY_CLASSIFICATION_V1_DRAFT[
                "artifact_write_status"
            ]
            ==
            "NONE"
        ),
}


cell_66_pass = all(
    cell66_gates.values()
)


# ============================================================
# 21) Output
# ============================================================

print(
    "=== EXERCISE V2 RESIDUAL DEPENDENCY CLASSIFICATION "
    "— RUN-ALL STABLE REPAIR ==="
)


print(
    "\n=== CELL65 AUTHORITATIVE UNIVERSE ==="
)

print(
    "Residual candidates:",
    len(
        cell66_residual_candidates
    )
)

print(
    "Unique names:",
    len(
        cell66_residual_name_set
    )
)

print(
    "Names:",
    cell66_residual_names
)


print(
    "\n=== CLASSIFICATION COVERAGE ==="
)

print(
    "Missing:",
    cell66_missing_classification_names
)

print(
    "Extra:",
    cell66_extra_classification_names
)

print(
    "Runtime missing:",
    cell66_runtime_missing_objects
)

print(
    "Unresolved:",
    [
        row["variable_name"]
        for row in cell66_unresolved_classifications
    ]
)


print(
    "\n=== CLASSIFICATION COUNTS ==="
)

for classification_name in [
    "DERIVED_REBUILD",
    "HISTORICAL_PRESERVE",
    "SAFE_NO_ACTION",
    "CURRENT_STATE_REPLACE",
]:

    print(
        "-",
        classification_name,
        ":",
        cell66_classification_counts.get(
            classification_name,
            0,
        ),
    )


print(
    "\n=== TRUE DERIVED REBUILD — MUST BE 5 ==="
)

for name in cell66_derived_rebuild_objects:
    print("-", name)


print(
    "Missing derived:",
    cell66_missing_derived_rebuilds
)

print(
    "Unexpected derived:",
    cell66_unexpected_derived_rebuilds
)


print(
    "\n=== CORRECTED HISTORICAL OBJECTS ==="
)

for name in sorted(
    CELL66_CORRECTED_HISTORICAL_OBJECTS
):
    print(
        "-",
        name,
        "| present in current universe:",
        name in cell66_residual_name_set,
        "| classified historical:",
        name in cell66_historical_preserve_objects,
    )


print(
    "\n=== HISTORICAL PRESERVE ==="
)

for name in cell66_historical_preserve_objects:
    print("-", name)


print(
    "\n=== SAFE NO ACTION ==="
)

for name in cell66_safe_no_action_objects:
    print("-", name)


print(
    "\n=== CANONICAL INDEX REBUILDS ==="
)

for name in cell66_canonical_index_rebuilds:
    print("-", name)


print(
    "\nHistorical Legacy reference count:",
    cell66_historical_reference_total
)


print(
    "\nSource diagnostic:"
)

print(
    "Found:",
    len(
        cell66_objects_with_source_occurrence
    )
)

print(
    "Not found:",
    cell66_objects_without_source_occurrence
)


print(
    "\nSTATUS:",
    cell66_dependency_classification_status
)


print(
    "\n=== CODE CELL 66 GATES ==="
)

for gate_name, passed in cell66_gates.items():

    print(
        gate_name,
        ":",
        "PASS"
        if passed
        else "FAIL"
    )


print(
    "\nFAILED GATES:",
    [
        gate_name

        for gate_name, passed
        in cell66_gates.items()

        if not passed
    ]
)


print(
    "\nCODE CELL 66 GATE:",
    "PASS"
    if cell_66_pass
    else "FAIL"
)


assert cell_66_pass, (
    "Residual dependency classification is still invalid. "
    "Do not advance beyond Cell 66. Use the printed exact "
    "Cell65 universe and classification diagnostics."
)
```

<h3>Atomic Activation Transaction Bundle and Rollback Validation</h3>

<p>
Cell 65 ثابت کرد که Legacy Exercise Slice را می‌توان بدون تغییر Countها و بدون آسیب به
Non-Exercise State با V2 Slice جایگزین کرد.
Cell 66 نیز تمام Dependencyهای باقی‌مانده را طبقه‌بندی کرد.
</p>

<p>
اکنون قبل از هر Activation واقعی، باید یک Transaction دقیق و برگشت‌پذیر تعریف شود.
</p>

<p>
این Cell سه State مستقل می‌سازد:
</p>

<ul>
<li><strong>BEFORE</strong> — Snapshot دقیق Current Active State؛</li>
<li><strong>AFTER</strong> — همان Simulated V2 State که در Cell 65 Validation شده است؛</li>
<li><strong>ROLLBACK</strong> — Clone مستقل از BEFORE برای برگشت کامل در صورت Failure.</li>
</ul>

<p>
همچنین چهار Index Canonical که Cell 65 و Cell 66 مشخص کردند، روی AFTER از صفر بازسازی می‌شوند:
</p>

<ul>
<li><code>golden_case_records_by_id</code></li>
<li><code>task_draft_plans</code></li>
<li><code>target_evidence_by_case_id</code></li>
<li><code>target_evidence_by_task_draft_id</code></li>
</ul>

<p>
در این مرحله فقط 5 Current-State Derived Runtime Object در Transaction Manifest
به‌عنوان الزامات Post-Activation Rebuild ثبت می‌شوند. 13 Historical Object عمداً حفظ می‌شوند و
8 Scratch Object هیچ Transaction Action دریافت نمی‌کنند. Activation واقعی تا Validation مستقل
Rebuild همان 5 Object همچنان Blocked باقی می‌ماند.
</p>

<p>
هیچ Active Registry در این Cell تغییر نمی‌کند.
</p>

</div>

```python
# Code Cell 67

# ============================================================
# Atomic Exercise V2 Activation Transaction Bundle
#
# PURPOSE:
#
#   Construct a completely isolated BEFORE / AFTER / ROLLBACK
#   transaction model for the eventual Exercise V2 activation.
#
# BEFORE:
#   exact current canonical active state.
#
# AFTER:
#   Cell65 validated simulated state + rebuilt canonical indexes.
#
# ROLLBACK:
#   independent deep clone of BEFORE.
#
# This Cell DOES NOT:
#   - mutate active registries;
#   - execute activation;
#   - rebuild the 8 derived runtime objects;
#   - generate questions;
#   - write artifacts.
#
# ============================================================


# ============================================================
# 1) Helpers
# ============================================================

def cell67_clone(
    value,
):

    return cell52_safe_clone(
        value
    )


def cell67_sha(
    value,
):

    return hashlib.sha256(
        stable_json_key(
            value
        ).encode(
            "utf-8"
        )
    ).hexdigest()


def cell67_rows(
    value,
):

    if isinstance(
        value,
        dict,
    ):

        return list(
            value.values()
        )


    return list(
        value
    )


# ============================================================
# 2) Preconditions from Cells 65 and 66
# ============================================================

cell67_prerequisite_failures = []


if not (
    cell65_structural_simulation_pass
):

    cell67_prerequisite_failures.append(
        "CELL65_STRUCTURAL_SIMULATION_NOT_PASS"
    )


if (
    cell65_activation_readiness_status
    !=
    "CORE_SIMULATION_PASS_"
    "PENDING_RESIDUAL_DEPENDENCY_CLASSIFICATION"
):

    cell67_prerequisite_failures.append(
        "CELL65_ACTIVATION_READINESS_STATUS_UNEXPECTED"
    )


if (
    cell66_dependency_classification_status
    !=
    "ALL_RESIDUAL_DEPENDENCIES_CLASSIFIED"
):

    cell67_prerequisite_failures.append(
        "CELL66_DEPENDENCY_CLASSIFICATION_NOT_COMPLETE"
    )


if (
    len(
        cell66_derived_rebuild_objects
    )
    !=
    5
):

    cell67_prerequisite_failures.append(
        "EXPECTED_5_TRUE_DERIVED_REBUILDS_NOT_RESOLVED"
    )


if (
    len(
        cell66_current_state_replace_objects
    )
    !=
    0
):

    cell67_prerequisite_failures.append(
        "UNEXPECTED_EXTRA_CURRENT_STATE_REPLACEMENTS"
    )


# ============================================================
# 3) Freeze exact ACTIVE state at transaction-planning time
# ============================================================

cell67_active_state_before = {
    "golden_case_records":
        cell67_clone(
            golden_case_records
        ),

    "auto_case_records":
        cell67_clone(
            auto_case_records
        ),

    "case_groundability_precheck":
        cell67_clone(
            case_groundability_precheck
        ),

    "case_task_design":
        cell67_clone(
            case_task_design
        ),

    "task_draft_plan_rows":
        cell67_clone(
            task_draft_plan_rows
        ),

    "target_evidence_records":
        cell67_clone(
            target_evidence_records
        ),

    "pilot_selection_rows":
        cell67_clone(
            pilot_selection_rows
        ),
}


cell67_active_indexes_before = {
    "golden_case_records_by_id":
        cell67_clone(
            golden_case_records_by_id
        ),

    "task_draft_plans":
        cell67_clone(
            task_draft_plans
        ),

    "target_evidence_by_case_id":
        cell67_clone(
            target_evidence_by_case_id
        ),

    "target_evidence_by_task_draft_id":
        cell67_clone(
            target_evidence_by_task_draft_id
        ),
}


# ============================================================
# 4) BEFORE fingerprints
# ============================================================

cell67_before_registry_fingerprints = {
    name:
        cell67_sha(
            value
        )

    for name, value
    in cell67_active_state_before.items()
}


cell67_before_index_fingerprints = {
    name:
        cell67_sha(
            value
        )

    for name, value
    in cell67_active_indexes_before.items()
}


cell67_before_bundle_fingerprint = (
    cell67_sha(
        {
            "registries":
                cell67_active_state_before,

            "indexes":
                cell67_active_indexes_before,
        }
    )
)


# ============================================================
# 5) Build isolated AFTER canonical registries
#
# Authoritative source:
#   Cell65 validated simulated activation state.
# ============================================================

cell67_after_state = {
    "golden_case_records":
        cell67_clone(
            cell65_simulated_golden_case_records
        ),

    "auto_case_records":
        cell67_clone(
            cell65_simulated_auto_case_records
        ),

    "case_groundability_precheck":
        cell67_clone(
            cell65_simulated_groundability_rows
        ),

    "case_task_design":
        cell67_clone(
            cell65_simulated_task_design_rows
        ),

    "task_draft_plan_rows":
        cell67_clone(
            cell65_simulated_task_draft_plan_rows
        ),

    "target_evidence_records":
        cell67_clone(
            cell65_simulated_target_evidence_records
        ),

    "pilot_selection_rows":
        cell67_clone(
            cell65_simulated_pilot_selection_rows
        ),
}


# ============================================================
# 6) Rebuild four canonical indexes FROM AFTER state
#
# Do not reuse Cell65 index objects here.
# Recompute independently.
# ============================================================

cell67_after_golden_rows = (
    cell67_rows(
        cell67_after_state[
            "golden_case_records"
        ]
    )
)


cell67_after_task_rows = (
    cell67_rows(
        cell67_after_state[
            "task_draft_plan_rows"
        ]
    )
)


cell67_after_target_rows = (
    cell67_rows(
        cell67_after_state[
            "target_evidence_records"
        ]
    )
)


cell67_after_indexes = {
    "golden_case_records_by_id": {
        row[
            "golden_case_id"
        ]:
            row

        for row
        in cell67_after_golden_rows
    },

    "task_draft_plans": {
        row[
            "task_draft_id"
        ]:
            row

        for row
        in cell67_after_task_rows
    },

    "target_evidence_by_case_id": {
        row[
            "golden_case_id"
        ]:
            row

        for row
        in cell67_after_target_rows
    },

    "target_evidence_by_task_draft_id": {
        row[
            "task_draft_id"
        ]:
            row

        for row
        in cell67_after_target_rows
    },
}


# ============================================================
# 7) Independent index cardinality / linkage validation
# ============================================================

cell67_after_index_failures = []


if (
    len(
        cell67_after_indexes[
            "golden_case_records_by_id"
        ]
    )
    !=
    len(
        cell67_after_golden_rows
    )
):

    cell67_after_index_failures.append(
        "GOLDEN_CASE_INDEX_CARDINALITY_MISMATCH"
    )


if (
    len(
        cell67_after_indexes[
            "task_draft_plans"
        ]
    )
    !=
    len(
        cell67_after_task_rows
    )
):

    cell67_after_index_failures.append(
        "TASK_INDEX_CARDINALITY_MISMATCH"
    )


if (
    len(
        cell67_after_indexes[
            "target_evidence_by_case_id"
        ]
    )
    !=
    len(
        cell67_after_target_rows
    )
):

    cell67_after_index_failures.append(
        "TARGET_CASE_INDEX_CARDINALITY_MISMATCH"
    )


if (
    len(
        cell67_after_indexes[
            "target_evidence_by_task_draft_id"
        ]
    )
    !=
    len(
        cell67_after_target_rows
    )
):

    cell67_after_index_failures.append(
        "TARGET_TASK_INDEX_CARDINALITY_MISMATCH"
    )


for target_row in (
    cell67_after_target_rows
):

    case_id = (
        target_row[
            "golden_case_id"
        ]
    )


    task_id = (
        target_row[
            "task_draft_id"
        ]
    )


    target_id = (
        target_row[
            "target_evidence_id"
        ]
    )


    if (
        case_id
        not in
        cell67_after_indexes[
            "golden_case_records_by_id"
        ]
    ):

        cell67_after_index_failures.append(
            {
                "target_evidence_id":
                    target_id,

                "reason":
                    "CASE_MISSING",
            }
        )


    if (
        task_id
        not in
        cell67_after_indexes[
            "task_draft_plans"
        ]
    ):

        cell67_after_index_failures.append(
            {
                "target_evidence_id":
                    target_id,

                "reason":
                    "TASK_MISSING",
            }
        )

        continue


    if (
        cell67_after_indexes[
            "task_draft_plans"
        ][
            task_id
        ][
            "golden_case_id"
        ]
        !=
        case_id
    ):

        cell67_after_index_failures.append(
            {
                "target_evidence_id":
                    target_id,

                "reason":
                    "TASK_CASE_LINKAGE_MISMATCH",
            }
        )


# ============================================================
# 8) AFTER fingerprints
# ============================================================

cell67_after_registry_fingerprints = {
    name:
        cell67_sha(
            value
        )

    for name, value
    in cell67_after_state.items()
}


cell67_after_index_fingerprints = {
    name:
        cell67_sha(
            value
        )

    for name, value
    in cell67_after_indexes.items()
}


cell67_after_bundle_fingerprint = (
    cell67_sha(
        {
            "registries":
                cell67_after_state,

            "indexes":
                cell67_after_indexes,
        }
    )
)


# ============================================================
# 9) Prove AFTER canonical registries equal Cell65 validated
# simulated activation content
# ============================================================

cell67_after_vs_cell65_failures = []


CELL67_CELL65_EXPECTED_STATE = {
    "golden_case_records":
        cell65_simulated_golden_case_records,

    "auto_case_records":
        cell65_simulated_auto_case_records,

    "case_groundability_precheck":
        cell65_simulated_groundability_rows,

    "case_task_design":
        cell65_simulated_task_design_rows,

    "task_draft_plan_rows":
        cell65_simulated_task_draft_plan_rows,

    "target_evidence_records":
        cell65_simulated_target_evidence_records,

    "pilot_selection_rows":
        cell65_simulated_pilot_selection_rows,
}


for registry_name, expected_value in (
    CELL67_CELL65_EXPECTED_STATE.items()
):

    if (
        cell67_sha(
            cell67_after_state[
                registry_name
            ]
        )
        !=
        cell67_sha(
            expected_value
        )
    ):

        cell67_after_vs_cell65_failures.append(
            registry_name
        )


# ============================================================
# 10) Exact expected AFTER counts
# ============================================================

cell67_after_counts = {
    "golden_case_records":
        len(
            cell67_after_golden_rows
        ),

    "auto_case_records":
        len(
            cell67_rows(
                cell67_after_state[
                    "auto_case_records"
                ]
            )
        ),

    "case_groundability_precheck":
        len(
            cell67_rows(
                cell67_after_state[
                    "case_groundability_precheck"
                ]
            )
        ),

    "case_task_design":
        len(
            cell67_rows(
                cell67_after_state[
                    "case_task_design"
                ]
            )
        ),

    "task_draft_plan_rows":
        len(
            cell67_after_task_rows
        ),

    "target_evidence_records":
        len(
            cell67_after_target_rows
        ),

    "pilot_selection_rows":
        len(
            cell67_rows(
                cell67_after_state[
                    "pilot_selection_rows"
                ]
            )
        ),
}


CELL67_EXPECTED_AFTER_COUNTS = {
    "golden_case_records":
        724,

    "auto_case_records":
        572,

    "case_groundability_precheck":
        724,

    "case_task_design":
        724,

    "task_draft_plan_rows":
        724,

    "target_evidence_records":
        572,

    "pilot_selection_rows":
        23,
}


cell67_after_count_failures = [
    {
        "registry":
            registry_name,

        "expected":
            expected_count,

        "actual":
            cell67_after_counts[
                registry_name
            ],
    }

    for registry_name, expected_count
    in CELL67_EXPECTED_AFTER_COUNTS.items()

    if (
        cell67_after_counts[
            registry_name
        ]
        !=
        expected_count
    )
]


# ============================================================
# 11) Verify exact Legacy removal from AFTER
# ============================================================

cell67_after_serialized = (
    stable_json_key(
        {
            "registries":
                cell67_after_state,

            "indexes":
                cell67_after_indexes,
        }
    )
)


cell67_legacy_id_leakage = {
    "case_ids":
        sorted(
            [
                legacy_id

                for legacy_id
                in cell65_legacy_exercise_case_ids

                if (
                    legacy_id
                    in
                    cell67_after_serialized
                )
            ]
        ),

    "task_ids":
        sorted(
            [
                legacy_id

                for legacy_id
                in cell65_legacy_exercise_task_ids

                if (
                    legacy_id
                    in
                    cell67_after_serialized
                )
            ]
        ),

    "target_ids":
        sorted(
            [
                legacy_id

                for legacy_id
                in cell65_legacy_exercise_target_ids

                if (
                    legacy_id
                    in
                    cell67_after_serialized
                )
            ]
        ),
}


# ============================================================
# 12) Verify exact V2 admission into AFTER
# ============================================================

cell67_v2_admission_failures = []


for case_id in (
    cell65_v2_case_ids
):

    if (
        case_id
        not in
        cell67_after_indexes[
            "golden_case_records_by_id"
        ]
    ):

        cell67_v2_admission_failures.append(
            {
                "family":
                    "CASE",

                "id":
                    case_id,
            }
        )


for task_id in (
    cell65_v2_task_ids
):

    if (
        task_id
        not in
        cell67_after_indexes[
            "task_draft_plans"
        ]
    ):

        cell67_v2_admission_failures.append(
            {
                "family":
                    "TASK",

                "id":
                    task_id,
            }
        )


cell67_after_target_ids = {
    row[
        "target_evidence_id"
    ]

    for row
    in cell67_after_target_rows
}


for target_id in (
    cell65_v2_target_ids
):

    if (
        target_id
        not in
        cell67_after_target_ids
    ):

        cell67_v2_admission_failures.append(
            {
                "family":
                    "TARGET",

                "id":
                    target_id,
            }
        )


# ============================================================
# 13) Build independent ROLLBACK bundle
#
# Must reproduce BEFORE exactly and remain disjoint from AFTER.
# ============================================================

cell67_rollback_state = (
    cell67_clone(
        cell67_active_state_before
    )
)


cell67_rollback_indexes = (
    cell67_clone(
        cell67_active_indexes_before
    )
)


cell67_rollback_bundle_fingerprint = (
    cell67_sha(
        {
            "registries":
                cell67_rollback_state,

            "indexes":
                cell67_rollback_indexes,
        }
    )
)


cell67_rollback_reproduction_failures = []


if (
    cell67_rollback_bundle_fingerprint
    !=
    cell67_before_bundle_fingerprint
):

    cell67_rollback_reproduction_failures.append(
        "ROLLBACK_BUNDLE_DOES_NOT_REPRODUCE_BEFORE"
    )


for registry_name in (
    cell67_active_state_before.keys()
):

    if (
        cell67_sha(
            cell67_rollback_state[
                registry_name
            ]
        )
        !=
        cell67_before_registry_fingerprints[
            registry_name
        ]
    ):

        cell67_rollback_reproduction_failures.append(
            {
                "registry":
                    registry_name,

                "reason":
                    "ROLLBACK_REGISTRY_MISMATCH",
            }
        )


for index_name in (
    cell67_active_indexes_before.keys()
):

    if (
        cell67_sha(
            cell67_rollback_indexes[
                index_name
            ]
        )
        !=
        cell67_before_index_fingerprints[
            index_name
        ]
    ):

        cell67_rollback_reproduction_failures.append(
            {
                "index":
                    index_name,

                "reason":
                    "ROLLBACK_INDEX_MISMATCH",
            }
        )


# ============================================================
# 14) BEFORE and AFTER must actually differ
#
# Otherwise there is no real transaction.
# ============================================================

cell67_changed_registry_names = sorted(
    [
        registry_name

        for registry_name
        in cell67_before_registry_fingerprints

        if (
            cell67_before_registry_fingerprints[
                registry_name
            ]
            !=
            cell67_after_registry_fingerprints[
                registry_name
            ]
        )
    ]
)


cell67_changed_index_names = sorted(
    [
        index_name

        for index_name
        in cell67_before_index_fingerprints

        if (
            cell67_before_index_fingerprints[
                index_name
            ]
            !=
            cell67_after_index_fingerprints[
                index_name
            ]
        )
    ]
)


# ============================================================
# 15) Derived runtime rebuild plan
#
# NOT executed here.
# ============================================================

cell67_derived_rebuild_plan = [
    {
        "variable_name":
            variable_name,

        "status":
            "NOT_EXECUTED",

        "execution_phase":
            "POST_CANONICAL_SWAP_PRE_QUESTION_GENERATION",
    }

    for variable_name
    in cell66_derived_rebuild_objects
]


cell67_canonical_index_rebuild_plan = [
    {
        "variable_name":
            variable_name,

        "status":
            "VALIDATED_IN_TRANSACTION_CLONE",
    }

    for variable_name
    in cell66_canonical_index_rebuilds
]


# ============================================================
# 16) Historical / scratch policy manifest
# ============================================================

cell67_historical_preserve_manifest = [
    {
        "variable_name":
            variable_name,

        "action":
            "PRESERVE_UNCHANGED",
    }

    for variable_name
    in cell66_historical_preserve_objects
]


cell67_safe_no_action_manifest = [
    {
        "variable_name":
            variable_name,

        "action":
            "NO_TRANSACTION_ACTION",
    }

    for variable_name
    in cell66_safe_no_action_objects
]


# ============================================================
# 17) Exact transaction phase plan
# ============================================================

CELL67_TRANSACTION_PHASES = [
    {
        "phase":
            1,

        "name":
            "PRECONDITION_FINGERPRINT_CHECK",

        "action":
            "VERIFY_ACTIVE_STATE_EQUALS_BEFORE_FINGERPRINT",
    },

    {
        "phase":
            2,

        "name":
            "CANONICAL_REGISTRY_SWAP",

        "action":
            "REPLACE_7_CANONICAL_REGISTRIES_WITH_VALIDATED_AFTER_STATE",
    },

    {
        "phase":
            3,

        "name":
            "CANONICAL_INDEX_REBUILD",

        "action":
            "REBUILD_4_CANONICAL_INDEXES_FROM_POST_SWAP_REGISTRIES",
    },

    {
        "phase":
            4,

        "name":
            "DERIVED_RUNTIME_REBUILD",

        "action":
            "REBUILD_8_CLASSIFIED_CURRENT_STATE_DERIVATIVES",
    },

    {
        "phase":
            5,

        "name":
            "POST_TRANSACTION_VALIDATION",

        "action":
            "VALIDATE_COUNTS_IDS_LINKAGE_PILOT_AND_DERIVED_STATE",
    },

    {
        "phase":
            6,

        "name":
            "COMMIT_OR_ROLLBACK",

        "action":
            "KEEP_AFTER_ONLY_IF_ALL_VALIDATION_PASSES_ELSE_RESTORE_BEFORE",
    },
]


# ============================================================
# 18) Transaction readiness
#
# Canonical transaction clone is validated.
#
# Actual activation remains blocked because the 8 derived
# rebuilds have not yet been independently reconstructed and
# tested against AFTER.
# ============================================================

cell67_transaction_clone_ready = (
    len(
        cell67_prerequisite_failures
    )
    ==
    0

    and

    len(
        cell67_after_index_failures
    )
    ==
    0

    and

    len(
        cell67_after_vs_cell65_failures
    )
    ==
    0

    and

    len(
        cell67_after_count_failures
    )
    ==
    0

    and

    len(
        cell67_legacy_id_leakage[
            "case_ids"
        ]
    )
    ==
    0

    and

    len(
        cell67_legacy_id_leakage[
            "task_ids"
        ]
    )
    ==
    0

    and

    len(
        cell67_legacy_id_leakage[
            "target_ids"
        ]
    )
    ==
    0

    and

    len(
        cell67_v2_admission_failures
    )
    ==
    0

    and

    len(
        cell67_rollback_reproduction_failures
    )
    ==
    0

    and

    cell67_before_bundle_fingerprint
    !=
    cell67_after_bundle_fingerprint

    and

    len(
        cell67_changed_registry_names
    )
    >
    0

    and

    len(
        cell67_changed_index_names
    )
    >
    0
)


cell67_activation_status = (
    "TRANSACTION_CLONE_VALIDATED_"
    "BLOCKED_PENDING_5_TRUE_DERIVED_REBUILDS"
    if (
        cell67_transaction_clone_ready
    )
    else
    "TRANSACTION_CLONE_VALIDATION_FAILED"
)


# ============================================================
# 19) Freeze transaction contract
# ============================================================

EXERCISE_V2_ATOMIC_TRANSACTION_PLAN_V1_DRAFT = {
    "version":
        "exercise_v2_atomic_transaction_plan_v1_draft",

    "status":
        cell67_activation_status,

    "before_bundle_fingerprint":
        cell67_before_bundle_fingerprint,

    "after_bundle_fingerprint":
        cell67_after_bundle_fingerprint,

    "rollback_bundle_fingerprint":
        cell67_rollback_bundle_fingerprint,

    "changed_canonical_registries":
        list(
            cell67_changed_registry_names
        ),

    "changed_canonical_indexes":
        list(
            cell67_changed_index_names
        ),

    "canonical_index_rebuilds":
        cell67_canonical_index_rebuild_plan,

    "derived_runtime_rebuilds":
        cell67_derived_rebuild_plan,

    "historical_preserve":
        cell67_historical_preserve_manifest,

    "safe_no_action":
        cell67_safe_no_action_manifest,

    "transaction_phases":
        cell67_clone(
            CELL67_TRANSACTION_PHASES
        ),

    "actual_activation_status":
        "NOT_STARTED",

    "derived_rebuild_execution_status":
        "NOT_STARTED",

    "rollback_execution_status":
        "NOT_NEEDED",

    "question_generation_status":
        "NOT_STARTED",

    "expected_answer_generation_status":
        "NOT_STARTED",

    "artifact_write_status":
        "NONE",
}


# ============================================================
# 20) Confirm ACTIVE state still equals BEFORE
# ============================================================

cell67_live_active_state_after_planning = {
    "golden_case_records":
        golden_case_records,

    "auto_case_records":
        auto_case_records,

    "case_groundability_precheck":
        case_groundability_precheck,

    "case_task_design":
        case_task_design,

    "task_draft_plan_rows":
        task_draft_plan_rows,

    "target_evidence_records":
        target_evidence_records,

    "pilot_selection_rows":
        pilot_selection_rows,
}


cell67_live_active_indexes_after_planning = {
    "golden_case_records_by_id":
        golden_case_records_by_id,

    "task_draft_plans":
        task_draft_plans,

    "target_evidence_by_case_id":
        target_evidence_by_case_id,

    "target_evidence_by_task_draft_id":
        target_evidence_by_task_draft_id,
}


cell67_live_mutation_failures = []


for registry_name, live_value in (
    cell67_live_active_state_after_planning.items()
):

    if (
        cell67_sha(
            live_value
        )
        !=
        cell67_before_registry_fingerprints[
            registry_name
        ]
    ):

        cell67_live_mutation_failures.append(
            {
                "registry":
                    registry_name,

                "reason":
                    "ACTIVE_REGISTRY_MUTATED_DURING_TRANSACTION_PLANNING",
            }
        )


for index_name, live_value in (
    cell67_live_active_indexes_after_planning.items()
):

    if (
        cell67_sha(
            live_value
        )
        !=
        cell67_before_index_fingerprints[
            index_name
        ]
    ):

        cell67_live_mutation_failures.append(
            {
                "index":
                    index_name,

                "reason":
                    "ACTIVE_INDEX_MUTATED_DURING_TRANSACTION_PLANNING",
            }
        )


# ============================================================
# 21) Gates
# ============================================================

cell67_gates = {
    "cell65_and_cell66_activation_preconditions_are_satisfied":
        (
            len(
                cell67_prerequisite_failures
            )
            ==
            0
        ),

    "before_state_contains_all_7_canonical_registries":
        (
            len(
                cell67_active_state_before
            )
            ==
            7
        ),

    "before_state_contains_all_4_canonical_indexes":
        (
            len(
                cell67_active_indexes_before
            )
            ==
            4
        ),

    "after_state_contains_all_7_canonical_registries":
        (
            len(
                cell67_after_state
            )
            ==
            7
        ),

    "all_4_after_indexes_rebuild_independently_without_failure":
        (
            len(
                cell67_after_indexes
            )
            ==
            4

            and

            len(
                cell67_after_index_failures
            )
            ==
            0
        ),

    "after_state_exactly_matches_cell65_validated_simulation":
        (
            len(
                cell67_after_vs_cell65_failures
            )
            ==
            0
        ),

    "after_state_preserves_expected_canonical_counts":
        (
            len(
                cell67_after_count_failures
            )
            ==
            0
        ),

    "no_legacy_exercise_case_task_target_id_survives_in_after_state":
        (
            len(
                cell67_legacy_id_leakage[
                    "case_ids"
                ]
            )
            ==
            0

            and

            len(
                cell67_legacy_id_leakage[
                    "task_ids"
                ]
            )
            ==
            0

            and

            len(
                cell67_legacy_id_leakage[
                    "target_ids"
                ]
            )
            ==
            0
        ),

    "all_v2_case_task_target_ids_are_present_in_after_state":
        (
            len(
                cell67_v2_admission_failures
            )
            ==
            0
        ),

    "rollback_bundle_reproduces_before_state_exactly":
        (
            len(
                cell67_rollback_reproduction_failures
            )
            ==
            0

            and

            cell67_rollback_bundle_fingerprint
            ==
            cell67_before_bundle_fingerprint
        ),

    "before_and_after_fingerprints_are_distinct":
        (
            cell67_before_bundle_fingerprint
            !=
            cell67_after_bundle_fingerprint
        ),

    "canonical_registries_and_indexes_have_real_transaction_changes":
        (
            len(
                cell67_changed_registry_names
            )
            >
            0

            and

            len(
                cell67_changed_index_names
            )
            >
            0
        ),

    "exact_4_canonical_index_rebuilds_are_in_transaction_plan":
        (
            len(
                cell67_canonical_index_rebuild_plan
            )
            ==
            4

            and

            {
                row[
                    "variable_name"
                ]

                for row
                in cell67_canonical_index_rebuild_plan
            }
            ==
            set(
                cell66_canonical_index_rebuilds
            )
        ),

    "exact_5_true_derived_runtime_rebuilds_are_deferred_but_explicit":
        (
            len(
                cell67_derived_rebuild_plan
            )
            ==
            5

            and

            {
                row[
                    "variable_name"
                ]

                for row
                in cell67_derived_rebuild_plan
            }
            ==
            set(
                cell66_derived_rebuild_objects
            )

            and

            all(
                row[
                    "status"
                ]
                ==
                "NOT_EXECUTED"

                for row
                in cell67_derived_rebuild_plan
            )
        ),

    "all_13_historical_objects_are_preserved_by_transaction_policy":
        (
            len(
                cell67_historical_preserve_manifest
            )
            ==
            13

            and

            all(
                row[
                    "action"
                ]
                ==
                "PRESERVE_UNCHANGED"

                for row
                in cell67_historical_preserve_manifest
            )
        ),

    "all_8_safe_scratch_objects_receive_no_transaction_action":
        (
            len(
                cell67_safe_no_action_manifest
            )
            ==
            8

            and

            all(
                row[
                    "action"
                ]
                ==
                "NO_TRANSACTION_ACTION"

                for row
                in cell67_safe_no_action_manifest
            )
        ),

    "transaction_clone_is_validated_but_real_activation_is_still_blocked":
        (
            cell67_activation_status
            ==
            "TRANSACTION_CLONE_VALIDATED_"
            "BLOCKED_PENDING_5_TRUE_DERIVED_REBUILDS"
        ),

    "live_active_state_and_indexes_remain_byte_stable":
        (
            len(
                cell67_live_mutation_failures
            )
            ==
            0
        ),

    "actual_activation_not_started":
        (
            EXERCISE_V2_ATOMIC_TRANSACTION_PLAN_V1_DRAFT[
                "actual_activation_status"
            ]
            ==
            "NOT_STARTED"
        ),

    "derived_rebuild_execution_not_started":
        (
            EXERCISE_V2_ATOMIC_TRANSACTION_PLAN_V1_DRAFT[
                "derived_rebuild_execution_status"
            ]
            ==
            "NOT_STARTED"
        ),

    "question_generation_not_started":
        (
            EXERCISE_V2_ATOMIC_TRANSACTION_PLAN_V1_DRAFT[
                "question_generation_status"
            ]
            ==
            "NOT_STARTED"
        ),

    "expected_answer_generation_not_started":
        (
            EXERCISE_V2_ATOMIC_TRANSACTION_PLAN_V1_DRAFT[
                "expected_answer_generation_status"
            ]
            ==
            "NOT_STARTED"
        ),

    "no_artifact_write":
        (
            EXERCISE_V2_ATOMIC_TRANSACTION_PLAN_V1_DRAFT[
                "artifact_write_status"
            ]
            ==
            "NONE"
        ),
}


cell_67_pass = all(
    cell67_gates.values()
)


# ============================================================
# 22) Output
# ============================================================

print(
    "=== EXERCISE V2 ATOMIC TRANSACTION BUNDLE ==="
)


print(
    "\n=== PRECONDITIONS ==="
)


print(
    "Failures:",
    cell67_prerequisite_failures
)


print(
    "\n=== BEFORE ==="
)


print(
    "Bundle SHA256:",
    cell67_before_bundle_fingerprint
)


for name, fingerprint in (
    cell67_before_registry_fingerprints.items()
):

    print(
        "- registry",
        name,
        ":",
        fingerprint,
    )


for name, fingerprint in (
    cell67_before_index_fingerprints.items()
):

    print(
        "- index",
        name,
        ":",
        fingerprint,
    )


print(
    "\n=== AFTER ==="
)


print(
    "Bundle SHA256:",
    cell67_after_bundle_fingerprint
)


print(
    "Counts:",
    cell67_after_counts
)


print(
    "Count failures:",
    cell67_after_count_failures
)


print(
    "Index failures:",
    cell67_after_index_failures
)


print(
    "Cell65 equivalence failures:",
    cell67_after_vs_cell65_failures
)


print(
    "Legacy ID leakage:",
    cell67_legacy_id_leakage
)


print(
    "V2 admission failures:",
    cell67_v2_admission_failures
)


print(
    "\nChanged canonical registries:"
)


for name in (
    cell67_changed_registry_names
):

    print(
        "-",
        name
    )


print(
    "\nChanged canonical indexes:"
)


for name in (
    cell67_changed_index_names
):

    print(
        "-",
        name
    )


print(
    "\n=== ROLLBACK ==="
)


print(
    "Rollback SHA256:",
    cell67_rollback_bundle_fingerprint
)


print(
    "Matches BEFORE:",
    (
        cell67_rollback_bundle_fingerprint
        ==
        cell67_before_bundle_fingerprint
    )
)


print(
    "Rollback failures:",
    cell67_rollback_reproduction_failures
)


print(
    "\n=== CANONICAL INDEX REBUILDS ==="
)


for row in (
    cell67_canonical_index_rebuild_plan
):

    print(
        "-",
        row
    )


print(
    "\n=== DEFERRED DERIVED RUNTIME REBUILDS ==="
)


for row in (
    cell67_derived_rebuild_plan
):

    print(
        "-",
        row
    )


print(
    "\n=== TRANSACTION PHASES ==="
)


for row in (
    CELL67_TRANSACTION_PHASES
):

    print(
        "-",
        row
    )


print(
    "\n=== TRANSACTION READINESS ==="
)


print(
    "Clone ready:",
    cell67_transaction_clone_ready
)


print(
    "Status:",
    cell67_activation_status
)


print(
    "Live mutation failures:",
    cell67_live_mutation_failures
)


print(
    "\n=== ACTIVE COUNTS — MUST REMAIN UNCHANGED ==="
)


print(
    "Golden:",
    len(
        golden_case_records
    )
)


print(
    "Auto:",
    len(
        cell65_active_auto_rows
    )
)


print(
    "Groundability:",
    len(
        cell65_active_groundability_rows
    )
)


print(
    "Task Design:",
    len(
        cell65_active_task_design_rows
    )
)


print(
    "Tasks:",
    len(
        task_draft_plan_rows
    )
)


print(
    "Targets:",
    len(
        target_evidence_records
    )
)


print(
    "Pilots:",
    len(
        pilot_selection_rows
    )
)


print(
    "\n=== CODE CELL 67 GATES ==="
)


for gate_name, passed in (
    cell67_gates.items()
):

    print(
        gate_name,
        ":",
        "PASS"
        if passed
        else "FAIL"
    )


print(
    "\nFAILED GATES:",
    [
        gate_name

        for gate_name, passed
        in cell67_gates.items()

        if not passed
    ]
)


print(
    "\nCODE CELL 67 GATE:",
    "PASS"
    if cell_67_pass
    else "FAIL"
)


assert cell_67_pass, (
    "Atomic Exercise V2 transaction bundle validation failed. "
    "Do not activate active registries. Repair this SAME Code "
    "Cell 67 until BEFORE, AFTER, rebuilt canonical indexes and "
    "ROLLBACK fingerprints all validate independently."
)
```

```python
# Code Cell 68

# ============================================================
# Exact Derived-Rebuild Contract Discovery
#
# Goal:
#   inspect the REAL Notebook definitions and runtime contracts
#   of the exact 8 DERIVED_REBUILD objects from Cell 66.
#
# IMPORTANT:
#   No formula is guessed here.
#   No derived object is rebuilt here.
#
# READ-ONLY.
#
# NO active mutation.
# NO activation.
# NO transaction execution.
# NO question generation.
# NO artifact write.
# ============================================================


# ============================================================
# 1) Helpers
# ============================================================

def cell68_sha(
    value,
):

    return hashlib.sha256(
        stable_json_key(
            value
        ).encode(
            "utf-8"
        )
    ).hexdigest()


def cell68_top_level_signature(
    value,
):

    if isinstance(
        value,
        dict,
    ):

        key_types = Counter(
            type(
                key
            ).__name__

            for key
            in value.keys()
        )


        value_types = Counter(
            type(
                item
            ).__name__

            for item
            in value.values()
        )


        return {
            "container_type":
                type(
                    value
                ).__name__,

            "length":
                len(
                    value
                ),

            "key_type_counts":
                dict(
                    key_types
                ),

            "value_type_counts":
                dict(
                    value_types
                ),
        }


    if isinstance(
        value,
        (
            list,
            tuple,
            set,
        ),
    ):

        item_types = Counter(
            type(
                item
            ).__name__

            for item
            in value
        )


        return {
            "container_type":
                type(
                    value
                ).__name__,

            "length":
                len(
                    value
                ),

            "item_type_counts":
                dict(
                    item_types
                ),
        }


    return {
        "container_type":
            type(
                value
            ).__name__,

        "length":
            None,
    }


# ============================================================
# 2) Snapshot Active + validated Transaction inputs
# ============================================================

cell68_active_snapshot_before = {
    "golden":
        cell68_sha(
            golden_case_records
        ),

    "auto":
        cell68_sha(
            auto_case_records
        ),

    "groundability":
        cell68_sha(
            case_groundability_precheck
        ),

    "task_design":
        cell68_sha(
            case_task_design
        ),

    "tasks":
        cell68_sha(
            task_draft_plan_rows
        ),

    "targets":
        cell68_sha(
            target_evidence_records
        ),

    "pilots":
        cell68_sha(
            pilot_selection_rows
        ),
}


cell68_transaction_after_snapshot_before = (
    cell68_sha(
        cell67_after_state
    )
)


cell68_transaction_before_snapshot_before = (
    cell68_sha(
        cell67_active_state_before
    )
)


# ============================================================
# 3) Eight-object dependency audit universe
#
# Cell66 already contains the corrected FINAL transaction policy
# (5 true derived / 13 historical / 8 safe). Cell68 retains the
# original eight candidate objects only as a source-contract audit
# so the three historical reclassifications remain independently
# evidenced.
# ============================================================

CELL68_EXPECTED_AUDIT_OBJECTS = {
    "candidate_hash_to_case_ids",
    "cross_stratum_repeated_candidate_hashes",
    "pilot_rows_by_token_size",
    "question_drafting_pilot_by_id",
    "question_drafting_pilot_packets",
    "ready_case_plans",
    "target_id_counts",
    "target_method_readiness",
}


# Backward-compatible variable names are retained because Cell69
# consumes the eight-object evidence map. They describe the AUDIT
# universe here, not the final transaction rebuild set.
CELL68_EXPECTED_DERIVED_OBJECTS = set(
    CELL68_EXPECTED_AUDIT_OBJECTS
)

cell68_derived_object_names = sorted(
    CELL68_EXPECTED_AUDIT_OBJECTS
)

cell68_derived_object_name_set = set(
    cell68_derived_object_names
)


# ============================================================
# 4) Runtime existence + exact fingerprints
# ============================================================

cell68_runtime_contracts = {}

cell68_runtime_missing = []


for variable_name in (
    cell68_derived_object_names
):

    if (
        variable_name
        not in
        globals()
    ):

        cell68_runtime_missing.append(
            variable_name
        )

        continue


    runtime_value = globals()[
        variable_name
    ]


    cell68_runtime_contracts[
        variable_name
    ] = {
        "type":
            type(
                runtime_value
            ).__name__,

        "top_level_signature":
            cell68_top_level_signature(
                runtime_value
            ),

        "sha256":
            cell68_sha(
                runtime_value
            ),
    }


# ============================================================
# 5) Legacy reference universe
# ============================================================

cell68_legacy_case_ids = set(
    cell65_legacy_exercise_case_ids
)


cell68_legacy_task_ids = set(
    cell65_legacy_exercise_task_ids
)


cell68_legacy_target_ids = set(
    cell65_legacy_exercise_target_ids
)


cell68_runtime_legacy_reference_profile = {}


for variable_name in (
    cell68_derived_object_names
):

    if (
        variable_name
        not in
        globals()
    ):

        continue


    serialized = stable_json_key(
        globals()[
            variable_name
        ]
    )


    matched_case_ids = sorted(
        [
            item_id

            for item_id
            in cell68_legacy_case_ids

            if (
                item_id
                in
                serialized
            )
        ]
    )


    matched_task_ids = sorted(
        [
            item_id

            for item_id
            in cell68_legacy_task_ids

            if (
                item_id
                in
                serialized
            )
        ]
    )


    matched_target_ids = sorted(
        [
            item_id

            for item_id
            in cell68_legacy_target_ids

            if (
                item_id
                in
                serialized
            )
        ]
    )


    cell68_runtime_legacy_reference_profile[
        variable_name
    ] = {
        "legacy_case_id_count":
            len(
                matched_case_ids
            ),

        "legacy_task_id_count":
            len(
                matched_task_ids
            ),

        "legacy_target_id_count":
            len(
                matched_target_ids
            ),

        "total_legacy_reference_identity_count":
            (
                len(
                    matched_case_ids
                )
                +
                len(
                    matched_task_ids
                )
                +
                len(
                    matched_target_ids
                )
            ),
    }


# ============================================================
# 6) Canonical dependency vocabulary
# ============================================================

CELL68_CANONICAL_DEPENDENCY_NAMES = {
    "golden_case_records",
    "golden_case_records_by_id",

    "auto_case_records",

    "case_groundability_precheck",

    "case_task_design",

    "task_draft_plan_rows",
    "task_draft_plans",

    "target_evidence_records",
    "target_evidence_by_case_id",
    "target_evidence_by_task_draft_id",

    "pilot_selection_rows",

    "targets_by_stratum",
}


# ============================================================
# 7) Search exact source occurrences
#
# Cell66 established that all 8 DERIVED_REBUILD objects were
# present in the older source snapshot; therefore source
# discovery is a valid gate for THESE eight objects.
# ============================================================

def cell68_source_occurrences(
    variable_name,
):

    occurrences = []


    for code_cell in (
        cell49_code_cells
    ):

        source = (
            code_cell[
                "source"
            ]
        )


        if (
            variable_name
            not in
            source
        ):

            continue


        source_lines = (
            source.splitlines()
        )


        for line_index, line in enumerate(
            source_lines
        ):

            if (
                variable_name
                not in
                line
            ):

                continue


            stripped = (
                line.strip()
            )


            probable_definition = (
                stripped.startswith(
                    variable_name
                    +
                    " "
                )

                or

                stripped.startswith(
                    variable_name
                    +
                    "="
                )

                or

                (
                    variable_name
                    +
                    " ="
                )
                in
                stripped

                or

                stripped.startswith(
                    variable_name
                    +
                    "["
                )

                or

                (
                    variable_name
                    +
                    ".append("
                )
                in
                stripped

                or

                (
                    variable_name
                    +
                    ".update("
                )
                in
                stripped
            )


            start_index = max(
                0,
                line_index - 20,
            )


            end_index = min(
                len(
                    source_lines
                ),
                line_index + 50,
            )


            snippet = "\n".join(
                source_lines[
                    start_index:
                    end_index
                ]
            )


            canonical_mentions = sorted(
                [
                    dependency_name

                    for dependency_name
                    in CELL68_CANONICAL_DEPENDENCY_NAMES

                    if (
                        dependency_name
                        in
                        snippet
                    )
                ]
            )


            occurrences.append(
                {
                    "cell_index":
                        code_cell[
                            "cell_index"
                        ],

                    "line_index":
                        line_index,

                    "line":
                        line,

                    "probable_definition":
                        probable_definition,

                    "canonical_mentions":
                        canonical_mentions,

                    "snippet":
                        snippet,
                }
            )


    return occurrences


cell68_source_evidence = {}


for variable_name in (
    cell68_derived_object_names
):

    occurrences = (
        cell68_source_occurrences(
            variable_name
        )
    )


    probable_definitions = [
        row

        for row
        in occurrences

        if (
            row[
                "probable_definition"
            ]
        )
    ]


    cell68_source_evidence[
        variable_name
    ] = {
        "occurrence_count":
            len(
                occurrences
            ),

        "definition_candidate_count":
            len(
                probable_definitions
            ),

        "occurrence_cells":
            sorted(
                {
                    row[
                        "cell_index"
                    ]

                    for row
                    in occurrences
                }
            ),

        "definition_candidate_cells":
            sorted(
                {
                    row[
                        "cell_index"
                    ]

                    for row
                    in probable_definitions
                }
            ),

        "all_canonical_mentions":
            sorted(
                set(
                    dependency_name

                    for row
                    in occurrences

                    for dependency_name
                    in row[
                        "canonical_mentions"
                    ]
                )
            ),

        "definition_candidates":
            probable_definitions,

        "all_occurrences":
            occurrences,
    }


# ============================================================
# 8) Source discovery failures
# ============================================================

cell68_source_discovery_failures = []


for variable_name in (
    cell68_derived_object_names
):

    evidence = (
        cell68_source_evidence[
            variable_name
        ]
    )


    if (
        evidence[
            "occurrence_count"
        ]
        ==
        0
    ):

        cell68_source_discovery_failures.append(
            {
                "variable_name":
                    variable_name,

                "reason":
                    "NO_SOURCE_OCCURRENCE",
            }
        )


    if (
        evidence[
            "definition_candidate_count"
        ]
        ==
        0
    ):

        cell68_source_discovery_failures.append(
            {
                "variable_name":
                    variable_name,

                "reason":
                    "NO_DEFINITION_CANDIDATE",
            }
        )


# ============================================================
# 9) Capture exact first definition candidate
#
# This does NOT claim the first occurrence alone is the whole
# formula. It freezes the source location from which the next
# Cell will reconstruct and independently reproduce the logic.
# ============================================================

cell68_primary_definition_evidence = {}


for variable_name in (
    cell68_derived_object_names
):

    definition_candidates = (
        cell68_source_evidence[
            variable_name
        ][
            "definition_candidates"
        ]
    )


    if (
        definition_candidates
    ):

        first_definition = (
            sorted(
                definition_candidates,
                key=lambda row:
                    (
                        row[
                            "cell_index"
                        ],
                        row[
                            "line_index"
                        ],
                    ),
            )[0]
        )


        cell68_primary_definition_evidence[
            variable_name
        ] = {
            "cell_index":
                first_definition[
                    "cell_index"
                ],

            "line_index":
                first_definition[
                    "line_index"
                ],

            "canonical_mentions":
                first_definition[
                    "canonical_mentions"
                ],

            "snippet":
                first_definition[
                    "snippet"
                ],
        }


# ============================================================
# 10) Runtime structural examples
#
# Keep output bounded.
# ============================================================

cell68_runtime_examples = {}


for variable_name in (
    cell68_derived_object_names
):

    if (
        variable_name
        not in
        globals()
    ):

        continue


    runtime_value = globals()[
        variable_name
    ]


    if isinstance(
        runtime_value,
        dict,
    ):

        example_items = list(
            runtime_value.items()
        )[
            :3
        ]


        cell68_runtime_examples[
            variable_name
        ] = [
            {
                "key":
                    cell52_safe_clone(
                        key
                    ),

                "value":
                    cell52_safe_clone(
                        value
                    ),
            }

            for key, value
            in example_items
        ]


    elif isinstance(
        runtime_value,
        (
            list,
            tuple,
        ),
    ):

        cell68_runtime_examples[
            variable_name
        ] = cell52_safe_clone(
            list(
                runtime_value
            )[
                :3
            ]
        )


    elif isinstance(
        runtime_value,
        set,
    ):

        cell68_runtime_examples[
            variable_name
        ] = cell52_safe_clone(
            sorted(
                list(
                    runtime_value
                ),
                key=lambda item:
                    stable_json_key(
                        item
                    ),
            )[
                :3
            ]
        )


    else:

        cell68_runtime_examples[
            variable_name
        ] = cell52_safe_clone(
            runtime_value
        )


# ============================================================
# 11) Pre-question criticality from Cell66
# ============================================================

CELL68_PREQUESTION_OBJECTS = set(
    CELL66_REQUIRED_PREQUESTION_REBUILDS
)


cell68_rebuild_phase_by_object = {}


for variable_name in (
    cell68_derived_object_names
):

    cell68_rebuild_phase_by_object[
        variable_name
    ] = (
        "POST_CANONICAL_SWAP_PRE_QUESTION_GENERATION"
        if (
            variable_name
            in
            CELL68_PREQUESTION_OBJECTS
        )
        else
        "POST_CANONICAL_SWAP_BEFORE_TRANSACTION_COMMIT"
    )


# ============================================================
# 12) Freeze exact discovery contract
# ============================================================

EXERCISE_V2_DERIVED_REBUILD_DISCOVERY_V1_DRAFT = {
    "version":
        "exercise_v2_derived_rebuild_discovery_v1_draft",

    "status":
        (
            "ALL_8_DEPENDENCY_AUDIT_CONTRACTS_DISCOVERED_"
            "ACTIVE_REPRODUCTION_NOT_STARTED"
            if (
                len(
                    cell68_runtime_missing
                )
                ==
                0

                and

                len(
                    cell68_source_discovery_failures
                )
                ==
                0

                and

                len(
                    cell68_primary_definition_evidence
                )
                ==
                8
            )
            else
            "DERIVED_CONTRACT_DISCOVERY_INCOMPLETE"
        ),

    "derived_object_count":
        len(
            cell68_derived_object_names
        ),

    "runtime_contracts":
        cell52_safe_clone(
            cell68_runtime_contracts
        ),

    "legacy_reference_profile":
        cell52_safe_clone(
            cell68_runtime_legacy_reference_profile
        ),

    "source_definition_locations": {
        variable_name: {
            "cell_index":
                row[
                    "cell_index"
                ],

            "line_index":
                row[
                    "line_index"
                ],

            "canonical_mentions":
                row[
                    "canonical_mentions"
                ],
        }

        for variable_name, row
        in cell68_primary_definition_evidence.items()
    },

    "rebuild_phase_by_object":
        cell52_safe_clone(
            cell68_rebuild_phase_by_object
        ),

    "active_reproduction_status":
        "NOT_STARTED",

    "after_clone_rebuild_status":
        "NOT_STARTED",

    "actual_activation_status":
        "NOT_STARTED",

    "question_generation_status":
        "NOT_STARTED",

    "artifact_write_status":
        "NONE",
}


# ============================================================
# 13) No-mutation recheck
# ============================================================

cell68_active_snapshot_after = {
    "golden":
        cell68_sha(
            golden_case_records
        ),

    "auto":
        cell68_sha(
            auto_case_records
        ),

    "groundability":
        cell68_sha(
            case_groundability_precheck
        ),

    "task_design":
        cell68_sha(
            case_task_design
        ),

    "tasks":
        cell68_sha(
            task_draft_plan_rows
        ),

    "targets":
        cell68_sha(
            target_evidence_records
        ),

    "pilots":
        cell68_sha(
            pilot_selection_rows
        ),
}


cell68_transaction_after_snapshot_after = (
    cell68_sha(
        cell67_after_state
    )
)


cell68_transaction_before_snapshot_after = (
    cell68_sha(
        cell67_active_state_before
    )
)


# ============================================================
# 14) Gates
# ============================================================

cell68_gates = {
    "cell67_transaction_clone_is_validated":
        (
            cell67_transaction_clone_ready
            is True

            and

            cell67_activation_status
            ==
            "TRANSACTION_CLONE_VALIDATED_"
            "BLOCKED_PENDING_5_TRUE_DERIVED_REBUILDS"
        ),

    "exact_8_dependency_audit_objects_are_resolved":
        (
            len(
                cell68_derived_object_names
            )
            ==
            8

            and

            cell68_derived_object_name_set
            ==
            CELL68_EXPECTED_AUDIT_OBJECTS

            and

            set(
                cell66_derived_rebuild_objects
            )
            ==
            {
                "candidate_hash_to_case_ids",
                "cross_stratum_repeated_candidate_hashes",
                "question_drafting_pilot_by_id",
                "question_drafting_pilot_packets",
                "target_id_counts",
            }
        ),

    "all_8_audit_runtime_objects_exist":
        (
            len(
                cell68_runtime_missing
            )
            ==
            0
        ),

    "all_8_audit_runtime_contracts_are_fingerprinted":
        (
            len(
                cell68_runtime_contracts
            )
            ==
            8

            and

            all(
                row[
                    "sha256"
                ]

                for row
                in cell68_runtime_contracts.values()
            )
        ),

    "all_8_audit_objects_retain_expected_legacy_dependency_surface":
        (
            len(
                cell68_runtime_legacy_reference_profile
            )
            ==
            8

            and

            all(
                row[
                    "total_legacy_reference_identity_count"
                ]
                >
                0

                for row
                in cell68_runtime_legacy_reference_profile.values()
            )
        ),

    "source_occurrences_and_definition_candidates_found_for_all_8_audit_objects":
        (
            len(
                cell68_source_discovery_failures
            )
            ==
            0
        ),

    "exact_primary_definition_location_frozen_for_all_8_audit_objects":
        (
            len(
                cell68_primary_definition_evidence
            )
            ==
            8
        ),

    "all_8_audit_dependency_phases_are_explicit":
        (
            len(
                cell68_rebuild_phase_by_object
            )
            ==
            8

            and

            all(
                value
                in
                {
                    "POST_CANONICAL_SWAP_PRE_QUESTION_GENERATION",
                    "POST_CANONICAL_SWAP_BEFORE_TRANSACTION_COMMIT",
                }

                for value
                in cell68_rebuild_phase_by_object.values()
            )
        ),

    "all_required_prequestion_dependencies_are_present":
        (
            CELL68_PREQUESTION_OBJECTS
            <=
            cell68_derived_object_name_set
        ),

    "discovery_contract_is_complete_but_reproduction_has_not_started":
        (
            EXERCISE_V2_DERIVED_REBUILD_DISCOVERY_V1_DRAFT[
                "status"
            ]
            ==
            "ALL_8_DEPENDENCY_AUDIT_CONTRACTS_DISCOVERED_"
            "ACTIVE_REPRODUCTION_NOT_STARTED"

            and

            EXERCISE_V2_DERIVED_REBUILD_DISCOVERY_V1_DRAFT[
                "active_reproduction_status"
            ]
            ==
            "NOT_STARTED"
        ),

    "active_state_remains_byte_stable":
        (
            cell68_active_snapshot_before
            ==
            cell68_active_snapshot_after
        ),

    "transaction_before_and_after_clones_remain_byte_stable":
        (
            cell68_transaction_before_snapshot_before
            ==
            cell68_transaction_before_snapshot_after

            and

            cell68_transaction_after_snapshot_before
            ==
            cell68_transaction_after_snapshot_after
        ),

    "actual_activation_not_started":
        (
            EXERCISE_V2_DERIVED_REBUILD_DISCOVERY_V1_DRAFT[
                "actual_activation_status"
            ]
            ==
            "NOT_STARTED"
        ),

    "question_generation_not_started":
        (
            EXERCISE_V2_DERIVED_REBUILD_DISCOVERY_V1_DRAFT[
                "question_generation_status"
            ]
            ==
            "NOT_STARTED"
        ),

    "no_artifact_write":
        (
            EXERCISE_V2_DERIVED_REBUILD_DISCOVERY_V1_DRAFT[
                "artifact_write_status"
            ]
            ==
            "NONE"
        ),
}


cell_68_pass = all(
    cell68_gates.values()
)


# ============================================================
# 15) Output
# ============================================================

print(
    "=== EXERCISE V2 DERIVED-REBUILD CONTRACT DISCOVERY ==="
)


print(
    "\n=== DERIVED OBJECT UNIVERSE ==="
)


print(
    "Objects:",
    len(
        cell68_derived_object_names
    )
)


for variable_name in (
    cell68_derived_object_names
):

    print(
        "-",
        variable_name
    )


print(
    "\nRuntime missing:",
    cell68_runtime_missing
)


print(
    "\n=== RUNTIME CONTRACTS ==="
)


for variable_name in (
    cell68_derived_object_names
):

    print(
        "\nOBJECT:",
        variable_name
    )


    print(
        "Runtime:",
        cell68_runtime_contracts.get(
            variable_name
        )
    )


    print(
        "Legacy refs:",
        cell68_runtime_legacy_reference_profile.get(
            variable_name
        )
    )


    print(
        "Rebuild phase:",
        cell68_rebuild_phase_by_object.get(
            variable_name
        )
    )


    print(
        "Example:",
        cell68_runtime_examples.get(
            variable_name
        )
    )


print(
    "\n=== SOURCE DISCOVERY ==="
)


print(
    "Discovery failures:",
    cell68_source_discovery_failures
)


for variable_name in (
    cell68_derived_object_names
):

    evidence = (
        cell68_source_evidence[
            variable_name
        ]
    )


    print(
        "\nOBJECT:",
        variable_name
    )


    print(
        "Occurrences:",
        evidence[
            "occurrence_count"
        ]
    )


    print(
        "Definition candidates:",
        evidence[
            "definition_candidate_count"
        ]
    )


    print(
        "Occurrence cells:",
        evidence[
            "occurrence_cells"
        ]
    )


    print(
        "Definition candidate cells:",
        evidence[
            "definition_candidate_cells"
        ]
    )


    print(
        "Canonical dependency mentions:",
        evidence[
            "all_canonical_mentions"
        ]
    )


    primary_definition = (
        cell68_primary_definition_evidence.get(
            variable_name
        )
    )


    if (
        primary_definition
        is not None
    ):

        print(
            "Primary definition location:",
            (
                primary_definition[
                    "cell_index"
                ],
                primary_definition[
                    "line_index"
                ],
            )
        )


        print(
            "PRIMARY SOURCE SNIPPET:"
        )


        print(
            primary_definition[
                "snippet"
            ]
        )


print(
    "\n=== DISCOVERY STATUS ==="
)


print(
    EXERCISE_V2_DERIVED_REBUILD_DISCOVERY_V1_DRAFT[
        "status"
    ]
)


print(
    "\n=== ACTIVE COUNTS — MUST REMAIN UNCHANGED ==="
)


print(
    "Golden:",
    len(
        golden_case_records
    )
)


print(
    "Auto:",
    len(
        cell65_active_auto_rows
    )
)


print(
    "Groundability:",
    len(
        cell65_active_groundability_rows
    )
)


print(
    "Task Design:",
    len(
        cell65_active_task_design_rows
    )
)


print(
    "Tasks:",
    len(
        task_draft_plan_rows
    )
)


print(
    "Targets:",
    len(
        target_evidence_records
    )
)


print(
    "Pilots:",
    len(
        pilot_selection_rows
    )
)


print(
    "\n=== CODE CELL 68 GATES ==="
)


for gate_name, passed in (
    cell68_gates.items()
):

    print(
        gate_name,
        ":",
        "PASS"
        if passed
        else "FAIL"
    )


print(
    "\nFAILED GATES:",
    [
        gate_name

        for gate_name, passed
        in cell68_gates.items()

        if not passed
    ]
)


print(
    "\nCODE CELL 68 GATE:",
    "PASS"
    if cell_68_pass
    else "FAIL"
)


assert cell_68_pass, (
    "Exact derived-rebuild contract discovery failed. "
    "Do not execute derived rebuilds and do not activate the "
    "Exercise V2 transaction. Repair this SAME Code Cell 68 "
    "until all 8 runtime objects and their actual Notebook "
    "definition sources are resolved."
)
```

```python
# Code Cell 70

# ============================================================
# Exact BEFORE Reproduction of the Five True Derived Objects
#
# True DERIVED_REBUILD set frozen by Cell 69:
#
#   candidate_hash_to_case_ids
#   cross_stratum_repeated_candidate_hashes
#   question_drafting_pilot_packets
#   question_drafting_pilot_by_id
#   target_id_counts
#
# Goal:
#   reproduce all five from BEFORE / current canonical state
#   and require SHA-for-SHA equality with runtime objects.
#
# Important:
#   The complex Question Drafting packet builder is NOT
#   manually reimplemented.
#
#   The actual Cell57 source build block is extracted and
#   executed in an isolated namespace.
#
# READ / REPRODUCE ONLY.
#
# NO runtime replacement.
# NO AFTER rebuild.
# NO activation.
# NO question generation.
# NO artifact write.
# ============================================================


# ============================================================
# 1) Helpers
# ============================================================

def cell70_sha(
    value,
):

    return hashlib.sha256(
        stable_json_key(
            value
        ).encode(
            "utf-8"
        )
    ).hexdigest()


def cell70_clone(
    value,
):

    return cell52_safe_clone(
        value
    )


# ============================================================
# 2) Preconditions
# ============================================================

CELL70_EXPECTED_TRUE_DERIVED = {
    "candidate_hash_to_case_ids",
    "cross_stratum_repeated_candidate_hashes",
    "question_drafting_pilot_packets",
    "question_drafting_pilot_by_id",
    "target_id_counts",
}


cell70_prerequisite_failures = []


if not (
    cell_69_pass
):

    cell70_prerequisite_failures.append(
        "CELL69_NOT_PASS"
    )


if (
    EXERCISE_V2_RESIDUAL_DEPENDENCY_CLASSIFICATION_V2_DRAFT[
        "status"
    ]
    !=
    "RECONCILED_5_DERIVED_REBUILDS"
):

    cell70_prerequisite_failures.append(
        "CELL69_V2_CLASSIFICATION_NOT_RECONCILED"
    )


if (
    set(
        cell69_v2_derived_rebuild_objects
    )
    !=
    CELL70_EXPECTED_TRUE_DERIVED
):

    cell70_prerequisite_failures.append(
        "TRUE_DERIVED_SET_NOT_EXACTLY_THE_EXPECTED_FIVE"
    )


if not (
    cell67_transaction_clone_ready
):

    cell70_prerequisite_failures.append(
        "CELL67_TRANSACTION_CLONE_NOT_READY"
    )


# ============================================================
# 3) Snapshot live canonical / derived / transaction state
# ============================================================

cell70_active_snapshot_before = {
    "golden":
        cell70_sha(
            golden_case_records
        ),

    "auto":
        cell70_sha(
            auto_case_records
        ),

    "groundability":
        cell70_sha(
            case_groundability_precheck
        ),

    "task_design":
        cell70_sha(
            case_task_design
        ),

    "tasks":
        cell70_sha(
            task_draft_plan_rows
        ),

    "targets":
        cell70_sha(
            target_evidence_records
        ),

    "pilots":
        cell70_sha(
            pilot_selection_rows
        ),
}


cell70_runtime_derived_snapshot_before = {
    variable_name:
        cell70_sha(
            globals()[
                variable_name
            ]
        )

    for variable_name
    in CELL70_EXPECTED_TRUE_DERIVED
}


cell70_transaction_before_sha_before = (
    cell70_sha(
        cell67_active_state_before
    )
)


cell70_transaction_after_sha_before = (
    cell70_sha(
        cell67_after_state
    )
)


# ============================================================
# 4) BEFORE canonical aliases
#
# Use transaction BEFORE snapshot rather than relying on
# potentially mutated aliases created by earlier Notebook
# cells.
# ============================================================

cell70_before_golden_case_records = (
    cell70_clone(
        cell67_active_state_before[
            "golden_case_records"
        ]
    )
)


cell70_before_task_draft_plan_rows = (
    cell70_clone(
        cell67_active_state_before[
            "task_draft_plan_rows"
        ]
    )
)


cell70_before_target_evidence_records = (
    cell70_clone(
        cell67_active_state_before[
            "target_evidence_records"
        ]
    )
)


cell70_before_pilot_selection_rows = (
    cell70_clone(
        cell67_active_state_before[
            "pilot_selection_rows"
        ]
    )
)


cell70_before_golden_case_records_by_id = (
    cell70_clone(
        cell67_active_indexes_before[
            "golden_case_records_by_id"
        ]
    )
)


cell70_before_task_draft_plans = (
    cell70_clone(
        cell67_active_indexes_before[
            "task_draft_plans"
        ]
    )
)


cell70_before_target_evidence_by_case_id = (
    cell70_clone(
        cell67_active_indexes_before[
            "target_evidence_by_case_id"
        ]
    )
)


cell70_before_target_evidence_by_task_draft_id = (
    cell70_clone(
        cell67_active_indexes_before[
            "target_evidence_by_task_draft_id"
        ]
    )
)


# ============================================================
# 5) Reproduce candidate_hash_to_case_ids EXACTLY from Source
# ============================================================

cell70_reproduced_candidate_hash_to_case_ids = defaultdict(
    list
)


for case_record in (
    cell70_before_golden_case_records
):

    cell70_reproduced_candidate_hash_to_case_ids[
        case_record[
            "candidate_identity_sha256"
        ]
    ].append(
        case_record[
            "golden_case_id"
        ]
    )


# ============================================================
# 6) Reproduce cross_stratum_repeated_candidate_hashes EXACTLY
# ============================================================

cell70_reproduced_cross_stratum_repeated_candidate_hashes = {
    candidate_hash:
        case_ids

    for candidate_hash, case_ids
    in cell70_reproduced_candidate_hash_to_case_ids.items()

    if (
        len(
            case_ids
        )
        >
        1
    )
}


# ============================================================
# 7) Reproduce target_id_counts EXACTLY from Source
# ============================================================

cell70_reproduced_target_id_counts = Counter(
    record[
        "target_evidence_id"
    ]

    for record
    in cell70_before_target_evidence_records
)


# ============================================================
# 8) Resolve exact Cell57 source
# ============================================================

cell70_cell57_source_matches = [
    code_cell[
        "source"
    ]

    for code_cell
    in cell49_code_cells

    if (
        code_cell[
            "cell_index"
        ]
        ==
        57
    )
]


cell70_source_resolution_failures = []


if (
    len(
        cell70_cell57_source_matches
    )
    !=
    1
):

    cell70_source_resolution_failures.append(
        {
            "reason":
                "CELL57_SOURCE_NOT_UNIQUE",

            "match_count":
                len(
                    cell70_cell57_source_matches
                ),
        }
    )


cell70_cell57_source = (
    cell70_cell57_source_matches[
        0
    ]

    if (
        len(
            cell70_cell57_source_matches
        )
        ==
        1
    )

    else ""
)


cell70_cell57_lines = (
    cell70_cell57_source.splitlines()
)


# ============================================================
# 9) Locate the exact Question Drafting build block
# ============================================================

cell70_packet_build_start_candidates = [
    line_index

    for line_index, line
    in enumerate(
        cell70_cell57_lines
    )

    if (
        line.strip()
        ==
        "question_drafting_pilot_packets = []"
    )
]


cell70_packet_build_start = (
    cell70_packet_build_start_candidates[
        0
    ]

    if (
        len(
            cell70_packet_build_start_candidates
        )
        ==
        1
    )

    else None
)


if (
    len(
        cell70_packet_build_start_candidates
    )
    !=
    1
):

    cell70_source_resolution_failures.append(
        {
            "reason":
                "QUESTION_PACKET_BUILD_START_NOT_UNIQUE",

            "matches":
                cell70_packet_build_start_candidates,
        }
    )


# Find the next numbered source section after Build Drafting
# Packets. Cell68 already identified this region as section 5.
cell70_packet_build_end_candidates = []


if (
    cell70_packet_build_start
    is not None
):

    for line_index in range(
        cell70_packet_build_start + 1,
        len(
            cell70_cell57_lines
        ),
    ):

        stripped = (
            cell70_cell57_lines[
                line_index
            ].strip()
        )


        if (
            stripped.startswith(
                "# 6)"
            )
        ):

            cell70_packet_build_end_candidates.append(
                line_index
            )

            break


cell70_packet_build_end = (
    cell70_packet_build_end_candidates[
        0
    ]

    if (
        len(
            cell70_packet_build_end_candidates
        )
        ==
        1
    )

    else None
)


if (
    len(
        cell70_packet_build_end_candidates
    )
    !=
    1
):

    cell70_source_resolution_failures.append(
        {
            "reason":
                "QUESTION_PACKET_BUILD_END_NOT_RESOLVED",

            "matches":
                cell70_packet_build_end_candidates,
        }
    )


# ============================================================
# 10) Extract exact executable source block
# ============================================================

cell70_packet_build_source = ""


if (
    cell70_packet_build_start
    is not None

    and

    cell70_packet_build_end
    is not None

    and

    cell70_packet_build_end
    >
    cell70_packet_build_start
):

    # Stop immediately before the separator belonging to
    # section 6 when possible.
    cell70_effective_end = (
        cell70_packet_build_end
    )


    while (
        cell70_effective_end
        >
        cell70_packet_build_start

        and

        cell70_cell57_lines[
            cell70_effective_end - 1
        ].strip()
        in
        {
            "",
            "# ============================================================",
        }
    ):

        cell70_effective_end -= 1


    cell70_packet_build_source = "\n".join(
        cell70_cell57_lines[
            cell70_packet_build_start:
            cell70_effective_end
        ]
    )


if not (
    cell70_packet_build_source.strip()
):

    cell70_source_resolution_failures.append(
        "QUESTION_PACKET_BUILD_SOURCE_EMPTY"
    )


# ============================================================
# 11) Source block semantic checks BEFORE execution
# ============================================================

CELL70_REQUIRED_PACKET_BLOCK_TERMS = {
    "question_drafting_pilot_packets",
    "question_drafting_pilot_by_id",
    "pilot_selection_rows",
    "golden_case_records_by_id",
    "task_draft_plan_rows",
    "target_evidence_by_task_draft_id",
}


cell70_missing_packet_block_terms = sorted(
    [
        term

        for term
        in CELL70_REQUIRED_PACKET_BLOCK_TERMS

        if (
            term
            not in
            cell70_packet_build_source
        )
    ]
)


if (
    cell70_missing_packet_block_terms
):

    cell70_source_resolution_failures.append(
        {
            "reason":
                "PACKET_BUILD_BLOCK_MISSING_REQUIRED_TERMS",

            "missing":
                cell70_missing_packet_block_terms,
        }
    )


# ============================================================
# 12) Construct isolated execution namespace
#
# Start with current Notebook namespace so all source-defined
# helpers/constants remain available.
#
# Then override canonical dependencies with BEFORE clones.
#
# Assignment inside the extracted block creates NEW local
# question_drafting_pilot_packets / by_id objects.
# ============================================================

cell70_packet_exec_namespace = dict(
    globals()
)


cell70_packet_exec_namespace.update(
    {
        "golden_case_records":
            cell70_clone(
                cell67_active_state_before[
                    "golden_case_records"
                ]
            ),

        "golden_case_records_by_id":
            cell70_clone(
                cell70_before_golden_case_records_by_id
            ),

        "auto_case_records":
            cell70_clone(
                cell67_active_state_before[
                    "auto_case_records"
                ]
            ),

        "case_groundability_precheck":
            cell70_clone(
                cell67_active_state_before[
                    "case_groundability_precheck"
                ]
            ),

        "case_task_design":
            cell70_clone(
                cell67_active_state_before[
                    "case_task_design"
                ]
            ),

        "task_draft_plan_rows":
            cell70_clone(
                cell70_before_task_draft_plan_rows
            ),

        "task_draft_plans":
            cell70_clone(
                cell70_before_task_draft_plans
            ),

        "target_evidence_records":
            cell70_clone(
                cell70_before_target_evidence_records
            ),

        "target_evidence_by_case_id":
            cell70_clone(
                cell70_before_target_evidence_by_case_id
            ),

        "target_evidence_by_task_draft_id":
            cell70_clone(
                cell70_before_target_evidence_by_task_draft_id
            ),

        "pilot_selection_rows":
            cell70_clone(
                cell70_before_pilot_selection_rows
            ),
    }
)


# Reconstruct targets_by_stratum from BEFORE as a defensive
# source-compatible dependency. The packet block may not need
# it directly, but the original Cell57 source environment did.
cell70_before_targets_by_stratum = defaultdict(
    list
)


for target_record in (
    cell70_before_target_evidence_records
):

    cell70_before_targets_by_stratum[
        (
            target_record[
                "source_type"
            ],
            target_record[
                "stratum"
            ],
        )
    ].append(
        target_record
    )


cell70_packet_exec_namespace[
    "targets_by_stratum"
] = (
    cell70_before_targets_by_stratum
)


# ============================================================
# 13) Execute exact Cell57 packet-build source in isolation
# ============================================================

cell70_packet_exec_failure = None


if not (
    cell70_source_resolution_failures
):

    try:

        exec(
            cell70_packet_build_source,
            cell70_packet_exec_namespace,
            cell70_packet_exec_namespace,
        )


    except Exception as exc:

        cell70_packet_exec_failure = {
            "error_type":
                type(
                    exc
                ).__name__,

            "error_message":
                str(
                    exc
                ),
        }


# ============================================================
# 14) Recover reproduced packet objects
# ============================================================

cell70_reproduced_question_drafting_pilot_packets = (
    cell70_packet_exec_namespace.get(
        "question_drafting_pilot_packets"
    )
)


cell70_reproduced_question_drafting_pilot_by_id = (
    cell70_packet_exec_namespace.get(
        "question_drafting_pilot_by_id"
    )
)


# ============================================================
# 15) Exact runtime comparison contracts
# ============================================================

cell70_reproduction_results = {}


cell70_reproduction_results[
    "candidate_hash_to_case_ids"
] = {
    "runtime_sha":
        cell70_sha(
            candidate_hash_to_case_ids
        ),

    "reproduced_sha":
        cell70_sha(
            cell70_reproduced_candidate_hash_to_case_ids
        ),

    "runtime_type":
        type(
            candidate_hash_to_case_ids
        ).__name__,

    "reproduced_type":
        type(
            cell70_reproduced_candidate_hash_to_case_ids
        ).__name__,

    "runtime_length":
        len(
            candidate_hash_to_case_ids
        ),

    "reproduced_length":
        len(
            cell70_reproduced_candidate_hash_to_case_ids
        ),
}


cell70_reproduction_results[
    "cross_stratum_repeated_candidate_hashes"
] = {
    "runtime_sha":
        cell70_sha(
            cross_stratum_repeated_candidate_hashes
        ),

    "reproduced_sha":
        cell70_sha(
            cell70_reproduced_cross_stratum_repeated_candidate_hashes
        ),

    "runtime_type":
        type(
            cross_stratum_repeated_candidate_hashes
        ).__name__,

    "reproduced_type":
        type(
            cell70_reproduced_cross_stratum_repeated_candidate_hashes
        ).__name__,

    "runtime_length":
        len(
            cross_stratum_repeated_candidate_hashes
        ),

    "reproduced_length":
        len(
            cell70_reproduced_cross_stratum_repeated_candidate_hashes
        ),
}


cell70_reproduction_results[
    "target_id_counts"
] = {
    "runtime_sha":
        cell70_sha(
            target_id_counts
        ),

    "reproduced_sha":
        cell70_sha(
            cell70_reproduced_target_id_counts
        ),

    "runtime_type":
        type(
            target_id_counts
        ).__name__,

    "reproduced_type":
        type(
            cell70_reproduced_target_id_counts
        ).__name__,

    "runtime_length":
        len(
            target_id_counts
        ),

    "reproduced_length":
        len(
            cell70_reproduced_target_id_counts
        ),
}


if (
    cell70_reproduced_question_drafting_pilot_packets
    is not None
):

    cell70_reproduction_results[
        "question_drafting_pilot_packets"
    ] = {
        "runtime_sha":
            cell70_sha(
                question_drafting_pilot_packets
            ),

        "reproduced_sha":
            cell70_sha(
                cell70_reproduced_question_drafting_pilot_packets
            ),

        "runtime_type":
            type(
                question_drafting_pilot_packets
            ).__name__,

        "reproduced_type":
            type(
                cell70_reproduced_question_drafting_pilot_packets
            ).__name__,

        "runtime_length":
            len(
                question_drafting_pilot_packets
            ),

        "reproduced_length":
            len(
                cell70_reproduced_question_drafting_pilot_packets
            ),
    }


if (
    cell70_reproduced_question_drafting_pilot_by_id
    is not None
):

    cell70_reproduction_results[
        "question_drafting_pilot_by_id"
    ] = {
        "runtime_sha":
            cell70_sha(
                question_drafting_pilot_by_id
            ),

        "reproduced_sha":
            cell70_sha(
                cell70_reproduced_question_drafting_pilot_by_id
            ),

        "runtime_type":
            type(
                question_drafting_pilot_by_id
            ).__name__,

        "reproduced_type":
            type(
                cell70_reproduced_question_drafting_pilot_by_id
            ).__name__,

        "runtime_length":
            len(
                question_drafting_pilot_by_id
            ),

        "reproduced_length":
            len(
                cell70_reproduced_question_drafting_pilot_by_id
            ),
    }


# ============================================================
# 16) Exact 5/5 SHA reproduction audit
# ============================================================

cell70_exact_reproduction_failures = []


for variable_name in sorted(
    CELL70_EXPECTED_TRUE_DERIVED
):

    result = (
        cell70_reproduction_results.get(
            variable_name
        )
    )


    if (
        result is None
    ):

        cell70_exact_reproduction_failures.append(
            {
                "variable_name":
                    variable_name,

                "reason":
                    "NO_REPRODUCTION_RESULT",
            }
        )

        continue


    if (
        result[
            "runtime_sha"
        ]
        !=
        result[
            "reproduced_sha"
        ]
    ):

        cell70_exact_reproduction_failures.append(
            {
                "variable_name":
                    variable_name,

                "reason":
                    "SHA_MISMATCH",

                "runtime_sha":
                    result[
                        "runtime_sha"
                    ],

                "reproduced_sha":
                    result[
                        "reproduced_sha"
                    ],

                "runtime_length":
                    result[
                        "runtime_length"
                    ],

                "reproduced_length":
                    result[
                        "reproduced_length"
                    ],
            }
        )


cell70_exact_reproduction_count = (
    len(
        CELL70_EXPECTED_TRUE_DERIVED
    )
    -
    len(
        cell70_exact_reproduction_failures
    )
)


# ============================================================
# 17) Exact packet schema / index relationship checks
# ============================================================

cell70_packet_structure_failures = []


if (
    cell70_reproduced_question_drafting_pilot_packets
    is not None
):

    if (
        len(
            cell70_reproduced_question_drafting_pilot_packets
        )
        !=
        23
    ):

        cell70_packet_structure_failures.append(
            "REPRODUCED_PACKET_COUNT_NOT_23"
        )


    runtime_packet_key_sets = {
        tuple(
            sorted(
                packet.keys()
            )
        )

        for packet
        in question_drafting_pilot_packets
    }


    reproduced_packet_key_sets = {
        tuple(
            sorted(
                packet.keys()
            )
        )

        for packet
        in cell70_reproduced_question_drafting_pilot_packets
    }


    if (
        runtime_packet_key_sets
        !=
        reproduced_packet_key_sets
    ):

        cell70_packet_structure_failures.append(
            "PACKET_SCHEMA_MISMATCH"
        )


if (
    cell70_reproduced_question_drafting_pilot_packets
    is not None

    and

    cell70_reproduced_question_drafting_pilot_by_id
    is not None
):

    independently_indexed_packets = {
        packet[
            "pilot_case_id"
        ]:
            packet

        for packet
        in cell70_reproduced_question_drafting_pilot_packets
    }


    if (
        cell70_sha(
            independently_indexed_packets
        )
        !=
        cell70_sha(
            cell70_reproduced_question_drafting_pilot_by_id
        )
    ):

        cell70_packet_structure_failures.append(
            "REPRODUCED_PACKET_INDEX_NOT_EXACT_PROJECTION_OF_PACKET_LIST"
        )


# ============================================================
# 18) Legacy surface should also reproduce exactly on BEFORE
# ============================================================

cell70_legacy_surface_reproduction_failures = []


for variable_name in (
    [
        "candidate_hash_to_case_ids",
        "cross_stratum_repeated_candidate_hashes",
        "question_drafting_pilot_packets",
        "question_drafting_pilot_by_id",
        "target_id_counts",
    ]
):

    reproduced_value = {
        "candidate_hash_to_case_ids":
            cell70_reproduced_candidate_hash_to_case_ids,

        "cross_stratum_repeated_candidate_hashes":
            cell70_reproduced_cross_stratum_repeated_candidate_hashes,

        "question_drafting_pilot_packets":
            cell70_reproduced_question_drafting_pilot_packets,

        "question_drafting_pilot_by_id":
            cell70_reproduced_question_drafting_pilot_by_id,

        "target_id_counts":
            cell70_reproduced_target_id_counts,
    }[
        variable_name
    ]


    if (
        reproduced_value
        is None
    ):

        continue


    serialized = stable_json_key(
        reproduced_value
    )


    reproduced_legacy_case_count = sum(
        1

        for item_id
        in cell65_legacy_exercise_case_ids

        if (
            item_id
            in
            serialized
        )
    )


    reproduced_legacy_task_count = sum(
        1

        for item_id
        in cell65_legacy_exercise_task_ids

        if (
            item_id
            in
            serialized
        )
    )


    reproduced_legacy_target_count = sum(
        1

        for item_id
        in cell65_legacy_exercise_target_ids

        if (
            item_id
            in
            serialized
        )
    )


    expected_profile = (
        cell68_runtime_legacy_reference_profile[
            variable_name
        ]
    )


    if not (
        reproduced_legacy_case_count
        ==
        expected_profile[
            "legacy_case_id_count"
        ]

        and

        reproduced_legacy_task_count
        ==
        expected_profile[
            "legacy_task_id_count"
        ]

        and

        reproduced_legacy_target_count
        ==
        expected_profile[
            "legacy_target_id_count"
        ]
    ):

        cell70_legacy_surface_reproduction_failures.append(
            {
                "variable_name":
                    variable_name,

                "reproduced":
                    {
                        "case":
                            reproduced_legacy_case_count,

                        "task":
                            reproduced_legacy_task_count,

                        "target":
                            reproduced_legacy_target_count,
                    },

                "expected":
                    expected_profile,
            }
        )


# ============================================================
# 19) Freeze active-reproduction contract
# ============================================================

EXERCISE_V2_TRUE_DERIVED_ACTIVE_REPRODUCTION_V1_DRAFT = {
    "version":
        "exercise_v2_true_derived_active_reproduction_v1_draft",

    "status":
        (
            "EXACT_BEFORE_REPRODUCTION_5_OF_5"
            if (
                len(
                    cell70_prerequisite_failures
                )
                ==
                0

                and

                len(
                    cell70_source_resolution_failures
                )
                ==
                0

                and

                cell70_packet_exec_failure
                is None

                and

                cell70_exact_reproduction_count
                ==
                5

                and

                len(
                    cell70_exact_reproduction_failures
                )
                ==
                0

                and

                len(
                    cell70_packet_structure_failures
                )
                ==
                0

                and

                len(
                    cell70_legacy_surface_reproduction_failures
                )
                ==
                0
            )
            else
            "BEFORE_REPRODUCTION_INCOMPLETE"
        ),

    "true_derived_objects":
        sorted(
            CELL70_EXPECTED_TRUE_DERIVED
        ),

    "exact_reproduction_count":
        cell70_exact_reproduction_count,

    "runtime_vs_reproduction":
        cell52_safe_clone(
            cell70_reproduction_results
        ),

    "question_packet_source": {
        "cell_index":
            57,

        "build_start_line":
            cell70_packet_build_start,

        "build_end_line":
            cell70_packet_build_end,

        "execution_method":
            "EXACT_SOURCE_BLOCK_IN_ISOLATED_NAMESPACE",
    },

    "after_clone_rebuild_status":
        "NOT_STARTED",

    "actual_activation_status":
        "NOT_STARTED",

    "question_generation_status":
        "NOT_STARTED",

    "artifact_write_status":
        "NONE",
}


# ============================================================
# 20) No-mutation recheck
# ============================================================

cell70_active_snapshot_after = {
    "golden":
        cell70_sha(
            golden_case_records
        ),

    "auto":
        cell70_sha(
            auto_case_records
        ),

    "groundability":
        cell70_sha(
            case_groundability_precheck
        ),

    "task_design":
        cell70_sha(
            case_task_design
        ),

    "tasks":
        cell70_sha(
            task_draft_plan_rows
        ),

    "targets":
        cell70_sha(
            target_evidence_records
        ),

    "pilots":
        cell70_sha(
            pilot_selection_rows
        ),
}


cell70_runtime_derived_snapshot_after = {
    variable_name:
        cell70_sha(
            globals()[
                variable_name
            ]
        )

    for variable_name
    in CELL70_EXPECTED_TRUE_DERIVED
}


cell70_transaction_before_sha_after = (
    cell70_sha(
        cell67_active_state_before
    )
)


cell70_transaction_after_sha_after = (
    cell70_sha(
        cell67_after_state
    )
)


# ============================================================
# 21) Gates
# ============================================================

cell70_gates = {
    "cell69_reconciled_five_object_contract_is_active":
        (
            len(
                cell70_prerequisite_failures
            )
            ==
            0
        ),

    "exact_cell57_question_packet_build_source_is_resolved":
        (
            len(
                cell70_source_resolution_failures
            )
            ==
            0

            and

            bool(
                cell70_packet_build_source.strip()
            )
        ),

    "exact_cell57_packet_block_executes_in_isolated_before_namespace":
        (
            cell70_packet_exec_failure
            is None
        ),

    "candidate_hash_to_case_ids_reproduces_sha_exactly":
        (
            "candidate_hash_to_case_ids"
            in
            cell70_reproduction_results

            and

            cell70_reproduction_results[
                "candidate_hash_to_case_ids"
            ][
                "runtime_sha"
            ]
            ==
            cell70_reproduction_results[
                "candidate_hash_to_case_ids"
            ][
                "reproduced_sha"
            ]
        ),

    "cross_stratum_repeated_candidate_hashes_reproduces_sha_exactly":
        (
            "cross_stratum_repeated_candidate_hashes"
            in
            cell70_reproduction_results

            and

            cell70_reproduction_results[
                "cross_stratum_repeated_candidate_hashes"
            ][
                "runtime_sha"
            ]
            ==
            cell70_reproduction_results[
                "cross_stratum_repeated_candidate_hashes"
            ][
                "reproduced_sha"
            ]
        ),

    "target_id_counts_reproduces_sha_exactly":
        (
            "target_id_counts"
            in
            cell70_reproduction_results

            and

            cell70_reproduction_results[
                "target_id_counts"
            ][
                "runtime_sha"
            ]
            ==
            cell70_reproduction_results[
                "target_id_counts"
            ][
                "reproduced_sha"
            ]
        ),

    "question_drafting_pilot_packets_reproduces_sha_exactly":
        (
            "question_drafting_pilot_packets"
            in
            cell70_reproduction_results

            and

            cell70_reproduction_results[
                "question_drafting_pilot_packets"
            ][
                "runtime_sha"
            ]
            ==
            cell70_reproduction_results[
                "question_drafting_pilot_packets"
            ][
                "reproduced_sha"
            ]
        ),

    "question_drafting_pilot_by_id_reproduces_sha_exactly":
        (
            "question_drafting_pilot_by_id"
            in
            cell70_reproduction_results

            and

            cell70_reproduction_results[
                "question_drafting_pilot_by_id"
            ][
                "runtime_sha"
            ]
            ==
            cell70_reproduction_results[
                "question_drafting_pilot_by_id"
            ][
                "reproduced_sha"
            ]
        ),

    "all_five_true_derived_objects_reproduce_sha_for_sha":
        (
            cell70_exact_reproduction_count
            ==
            5

            and

            len(
                cell70_exact_reproduction_failures
            )
            ==
            0
        ),

    "question_packet_schema_and_index_relationship_reproduce_exactly":
        (
            len(
                cell70_packet_structure_failures
            )
            ==
            0
        ),

    "legacy_dependency_surface_reproduces_exactly_on_before":
        (
            len(
                cell70_legacy_surface_reproduction_failures
            )
            ==
            0
        ),

    "active_reproduction_contract_is_exact_5_of_5":
        (
            EXERCISE_V2_TRUE_DERIVED_ACTIVE_REPRODUCTION_V1_DRAFT[
                "status"
            ]
            ==
            "EXACT_BEFORE_REPRODUCTION_5_OF_5"
        ),

    "all_live_canonical_registries_remain_byte_stable":
        (
            cell70_active_snapshot_before
            ==
            cell70_active_snapshot_after
        ),

    "all_five_live_derived_runtime_objects_remain_byte_stable":
        (
            cell70_runtime_derived_snapshot_before
            ==
            cell70_runtime_derived_snapshot_after
        ),

    "transaction_before_and_after_clones_remain_byte_stable":
        (
            cell70_transaction_before_sha_before
            ==
            cell70_transaction_before_sha_after

            and

            cell70_transaction_after_sha_before
            ==
            cell70_transaction_after_sha_after
        ),

    "after_clone_rebuild_not_started":
        (
            EXERCISE_V2_TRUE_DERIVED_ACTIVE_REPRODUCTION_V1_DRAFT[
                "after_clone_rebuild_status"
            ]
            ==
            "NOT_STARTED"
        ),

    "actual_activation_not_started":
        (
            EXERCISE_V2_TRUE_DERIVED_ACTIVE_REPRODUCTION_V1_DRAFT[
                "actual_activation_status"
            ]
            ==
            "NOT_STARTED"
        ),

    "question_generation_not_started":
        (
            EXERCISE_V2_TRUE_DERIVED_ACTIVE_REPRODUCTION_V1_DRAFT[
                "question_generation_status"
            ]
            ==
            "NOT_STARTED"
        ),

    "no_artifact_write":
        (
            EXERCISE_V2_TRUE_DERIVED_ACTIVE_REPRODUCTION_V1_DRAFT[
                "artifact_write_status"
            ]
            ==
            "NONE"
        ),
}


cell_70_pass = all(
    cell70_gates.values()
)


# ============================================================
# 22) Output
# ============================================================

print(
    "=== EXERCISE V2 TRUE-DERIVED BEFORE REPRODUCTION ==="
)


print(
    "\n=== PREREQUISITES ==="
)


print(
    "Failures:",
    cell70_prerequisite_failures
)


print(
    "\n=== CELL57 EXACT SOURCE BLOCK ==="
)


print(
    "Build start line:",
    cell70_packet_build_start
)


print(
    "Build end line:",
    cell70_packet_build_end
)


print(
    "Source resolution failures:",
    cell70_source_resolution_failures
)


print(
    "Packet exec failure:",
    cell70_packet_exec_failure
)


print(
    "\n=== 5 OBJECT SHA REPRODUCTION ==="
)


for variable_name in sorted(
    CELL70_EXPECTED_TRUE_DERIVED
):

    result = (
        cell70_reproduction_results.get(
            variable_name
        )
    )


    print(
        "\nOBJECT:",
        variable_name
    )


    if (
        result is None
    ):

        print(
            "NO RESULT"
        )

        continue


    print(
        "Runtime type:",
        result[
            "runtime_type"
        ]
    )


    print(
        "Reproduced type:",
        result[
            "reproduced_type"
        ]
    )


    print(
        "Runtime length:",
        result[
            "runtime_length"
        ]
    )


    print(
        "Reproduced length:",
        result[
            "reproduced_length"
        ]
    )


    print(
        "Runtime SHA:",
        result[
            "runtime_sha"
        ]
    )


    print(
        "Reproduced SHA:",
        result[
            "reproduced_sha"
        ]
    )


    print(
        "EXACT:",
        (
            result[
                "runtime_sha"
            ]
            ==
            result[
                "reproduced_sha"
            ]
        )
    )


print(
    "\nExact reproduction count:",
    cell70_exact_reproduction_count,
    "/ 5",
)


print(
    "Exact reproduction failures:",
    cell70_exact_reproduction_failures
)


print(
    "Packet structure failures:",
    cell70_packet_structure_failures
)


print(
    "Legacy surface reproduction failures:",
    cell70_legacy_surface_reproduction_failures
)


print(
    "\n=== ACTIVE COUNTS — MUST REMAIN UNCHANGED ==="
)


print(
    "Golden:",
    len(
        golden_case_records
    )
)


print(
    "Auto:",
    len(
        cell65_active_auto_rows
    )
)


print(
    "Groundability:",
    len(
        cell65_active_groundability_rows
    )
)


print(
    "Task Design:",
    len(
        cell65_active_task_design_rows
    )
)


print(
    "Tasks:",
    len(
        task_draft_plan_rows
    )
)


print(
    "Targets:",
    len(
        target_evidence_records
    )
)


print(
    "Pilots:",
    len(
        pilot_selection_rows
    )
)


print(
    "\n=== CODE CELL 70 GATES ==="
)


for gate_name, passed in (
    cell70_gates.items()
):

    print(
        gate_name,
        ":",
        "PASS"
        if passed
        else "FAIL"
    )


print(
    "\nFAILED GATES:",
    [
        gate_name

        for gate_name, passed
        in cell70_gates.items()

        if not passed
    ]
)


print(
    "\nCODE CELL 70 GATE:",
    "PASS"
    if cell_70_pass
    else "FAIL"
)


assert cell_70_pass, (
    "Exact BEFORE reproduction of the five true derived "
    "runtime objects failed. Do not rebuild them on AFTER and "
    "do not activate the Exercise V2 transaction. Repair this "
    "SAME Code Cell 70 until all five objects reproduce "
    "SHA-for-SHA from the current canonical BEFORE state."
)
```

```python
# Code Cell 71 — REPAIRED

# ============================================================
# Rebuild the Five True Derived Objects on AFTER Clone
#
# REPAIR:
#
# Previous Cell71 incorrectly treated:
#
#     golden_case_id
#
# as if it were:
#
#     pilot_case_id
#
# But Question Drafting packets have a distinct identity:
#
#     question-pilot:...
#
# Therefore Exercise packet transition must be resolved from
# actual packet records by (source_type, stratum) and then
# validated against the staged Case / Task / Target chain.
#
# Five true current-state derivatives:
#
#   candidate_hash_to_case_ids
#   cross_stratum_repeated_candidate_hashes
#   target_id_counts
#   question_drafting_pilot_packets
#   question_drafting_pilot_by_id
#
# Historical objects intentionally NOT rebuilt:
#
#   pilot_rows_by_token_size
#   ready_case_plans
#   target_method_readiness
#
# NO live mutation.
# NO actual activation.
# NO question generation.
# NO artifact write.
# ============================================================


# ============================================================
# 1) Helpers
# ============================================================

def cell71_sha(
    value,
):

    return hashlib.sha256(
        stable_json_key(
            value
        ).encode(
            "utf-8"
        )
    ).hexdigest()


def cell71_clone(
    value,
):

    return cell52_safe_clone(
        value
    )


def cell71_identity_keyed_sha(
    rows,
    key_field,
):

    keyed_rows = sorted(
        [
            {
                "key":
                    row[
                        key_field
                    ],

                "record":
                    row,
            }

            for row
            in rows
        ],
        key=lambda item:
            stable_json_key(
                item[
                    "key"
                ]
            ),
    )


    return cell71_sha(
        keyed_rows
    )


# ============================================================
# 2) Preconditions
# ============================================================

CELL71_TRUE_DERIVED_OBJECTS = {
    "candidate_hash_to_case_ids",
    "cross_stratum_repeated_candidate_hashes",
    "question_drafting_pilot_packets",
    "question_drafting_pilot_by_id",
    "target_id_counts",
}


CELL71_HISTORICAL_NOT_REBUILT = {
    "pilot_rows_by_token_size",
    "ready_case_plans",
    "target_method_readiness",
}


CELL71_EXERCISE_SOURCE_TYPE = (
    "book"
)


CELL71_EXERCISE_STRATUM = (
    "exercise_or_self_assessment"
)


cell71_prerequisite_failures = []


if not (
    cell_70_pass
):

    cell71_prerequisite_failures.append(
        "CELL70_NOT_PASS"
    )


if (
    EXERCISE_V2_TRUE_DERIVED_ACTIVE_REPRODUCTION_V1_DRAFT[
        "status"
    ]
    !=
    "EXACT_BEFORE_REPRODUCTION_5_OF_5"
):

    cell71_prerequisite_failures.append(
        "CELL70_EXACT_BEFORE_REPRODUCTION_NOT_5_OF_5"
    )


if (
    set(
        cell69_v2_derived_rebuild_objects
    )
    !=
    CELL71_TRUE_DERIVED_OBJECTS
):

    cell71_prerequisite_failures.append(
        "CELL69_TRUE_DERIVED_SET_MISMATCH"
    )


if not (
    CELL71_HISTORICAL_NOT_REBUILT
    <=
    set(
        cell69_v2_historical_preserve_objects
    )
):

    cell71_prerequisite_failures.append(
        "CELL69_HISTORICAL_RECLASSIFICATION_NOT_PRESERVED"
    )


if not (
    cell67_transaction_clone_ready
):

    cell71_prerequisite_failures.append(
        "CELL67_TRANSACTION_CLONE_NOT_READY"
    )


# ============================================================
# 3) Snapshot live canonical + runtime state
# ============================================================

cell71_live_active_snapshot_before = {
    "golden":
        cell71_sha(
            golden_case_records
        ),

    "auto":
        cell71_sha(
            auto_case_records
        ),

    "groundability":
        cell71_sha(
            case_groundability_precheck
        ),

    "task_design":
        cell71_sha(
            case_task_design
        ),

    "tasks":
        cell71_sha(
            task_draft_plan_rows
        ),

    "targets":
        cell71_sha(
            target_evidence_records
        ),

    "pilots":
        cell71_sha(
            pilot_selection_rows
        ),
}


cell71_live_true_derived_snapshot_before = {
    variable_name:
        cell71_sha(
            globals()[
                variable_name
            ]
        )

    for variable_name
    in CELL71_TRUE_DERIVED_OBJECTS
}


cell71_live_historical_snapshot_before = {
    variable_name:
        cell71_sha(
            globals()[
                variable_name
            ]
        )

    for variable_name
    in CELL71_HISTORICAL_NOT_REBUILT
}


cell71_transaction_before_sha_before = (
    cell71_sha(
        cell67_active_state_before
    )
)


cell71_transaction_after_sha_before = (
    cell71_sha(
        cell67_after_state
    )
)


cell71_transaction_after_indexes_sha_before = (
    cell71_sha(
        cell67_after_indexes
    )
)


# ============================================================
# 4) Isolated AFTER canonical aliases
# ============================================================

cell71_after_golden_case_records = (
    cell71_clone(
        cell67_after_state[
            "golden_case_records"
        ]
    )
)


cell71_after_auto_case_records = (
    cell71_clone(
        cell67_after_state[
            "auto_case_records"
        ]
    )
)


cell71_after_groundability = (
    cell71_clone(
        cell67_after_state[
            "case_groundability_precheck"
        ]
    )
)


cell71_after_task_design = (
    cell71_clone(
        cell67_after_state[
            "case_task_design"
        ]
    )
)


cell71_after_task_draft_plan_rows = (
    cell71_clone(
        cell67_after_state[
            "task_draft_plan_rows"
        ]
    )
)


cell71_after_target_evidence_records = (
    cell71_clone(
        cell67_after_state[
            "target_evidence_records"
        ]
    )
)


cell71_after_pilot_selection_rows = (
    cell71_clone(
        cell67_after_state[
            "pilot_selection_rows"
        ]
    )
)


cell71_after_golden_case_records_by_id = (
    cell71_clone(
        cell67_after_indexes[
            "golden_case_records_by_id"
        ]
    )
)


cell71_after_task_draft_plans = (
    cell71_clone(
        cell67_after_indexes[
            "task_draft_plans"
        ]
    )
)


cell71_after_target_evidence_by_case_id = (
    cell71_clone(
        cell67_after_indexes[
            "target_evidence_by_case_id"
        ]
    )
)


cell71_after_target_evidence_by_task_draft_id = (
    cell71_clone(
        cell67_after_indexes[
            "target_evidence_by_task_draft_id"
        ]
    )
)


# ============================================================
# 5) Rebuild candidate_hash_to_case_ids on AFTER
# ============================================================

cell71_after_candidate_hash_to_case_ids = defaultdict(
    list
)


for case_record in (
    cell71_after_golden_case_records
):

    cell71_after_candidate_hash_to_case_ids[
        case_record[
            "candidate_identity_sha256"
        ]
    ].append(
        case_record[
            "golden_case_id"
        ]
    )


# ============================================================
# 6) Rebuild cross_stratum_repeated_candidate_hashes
# ============================================================

cell71_after_cross_stratum_repeated_candidate_hashes = {
    candidate_hash:
        case_ids

    for candidate_hash, case_ids
    in cell71_after_candidate_hash_to_case_ids.items()

    if (
        len(
            case_ids
        )
        >
        1
    )
}


# ============================================================
# 7) Rebuild target_id_counts on AFTER
# ============================================================

cell71_after_target_id_counts = Counter(
    record[
        "target_evidence_id"
    ]

    for record
    in cell71_after_target_evidence_records
)


# ============================================================
# 8) Source-compatible targets_by_stratum
# ============================================================

cell71_after_targets_by_stratum = defaultdict(
    list
)


for target_record in (
    cell71_after_target_evidence_records
):

    cell71_after_targets_by_stratum[
        (
            target_record[
                "source_type"
            ],
            target_record[
                "stratum"
            ],
        )
    ].append(
        target_record
    )


# ============================================================
# 9) Isolated AFTER namespace for exact Cell57 builder
# ============================================================

cell71_packet_exec_namespace = dict(
    globals()
)


cell71_packet_exec_namespace.update(
    {
        "golden_case_records":
            cell71_after_golden_case_records,

        "golden_case_records_by_id":
            cell71_after_golden_case_records_by_id,

        "auto_case_records":
            cell71_after_auto_case_records,

        "case_groundability_precheck":
            cell71_after_groundability,

        "case_task_design":
            cell71_after_task_design,

        "task_draft_plan_rows":
            cell71_after_task_draft_plan_rows,

        "task_draft_plans":
            cell71_after_task_draft_plans,

        "target_evidence_records":
            cell71_after_target_evidence_records,

        "target_evidence_by_case_id":
            cell71_after_target_evidence_by_case_id,

        "target_evidence_by_task_draft_id":
            cell71_after_target_evidence_by_task_draft_id,

        "pilot_selection_rows":
            cell71_after_pilot_selection_rows,

        "targets_by_stratum":
            cell71_after_targets_by_stratum,
    }
)


# ============================================================
# 10) Execute exact proven Cell57 packet builder
# ============================================================

cell71_packet_exec_failure = None


try:

    exec(
        cell70_packet_build_source,
        cell71_packet_exec_namespace,
        cell71_packet_exec_namespace,
    )


except Exception as exc:

    cell71_packet_exec_failure = {
        "error_type":
            type(
                exc
            ).__name__,

        "error_message":
            str(
                exc
            ),
    }


# ============================================================
# 11) Recover AFTER packet objects
# ============================================================

cell71_after_question_drafting_pilot_packets = (
    cell71_packet_exec_namespace.get(
        "question_drafting_pilot_packets"
    )
)


cell71_after_question_drafting_pilot_by_id = (
    cell71_packet_exec_namespace.get(
        "question_drafting_pilot_by_id"
    )
)


# ============================================================
# 12) Complete isolated five-object AFTER bundle
# ============================================================

staged_exercise_v2_after_derived_runtime = {
    "candidate_hash_to_case_ids":
        cell71_after_candidate_hash_to_case_ids,

    "cross_stratum_repeated_candidate_hashes":
        cell71_after_cross_stratum_repeated_candidate_hashes,

    "question_drafting_pilot_packets":
        cell71_after_question_drafting_pilot_packets,

    "question_drafting_pilot_by_id":
        cell71_after_question_drafting_pilot_by_id,

    "target_id_counts":
        cell71_after_target_id_counts,
}


cell71_after_derived_fingerprints = {
    variable_name:
        (
            cell71_sha(
                value
            )
            if value is not None
            else None
        )

    for variable_name, value
    in staged_exercise_v2_after_derived_runtime.items()
}


cell71_after_derived_bundle_sha256 = (
    cell71_sha(
        staged_exercise_v2_after_derived_runtime
    )
)


# ============================================================
# 13) Candidate-hash validation
# ============================================================

cell71_candidate_hash_failures = []


cell71_candidate_hash_memberships = [
    case_id

    for case_ids
    in cell71_after_candidate_hash_to_case_ids.values()

    for case_id
    in case_ids
]


cell71_after_golden_case_ids = {
    row[
        "golden_case_id"
    ]

    for row
    in cell71_after_golden_case_records
}


if (
    len(
        cell71_candidate_hash_memberships
    )
    !=
    len(
        cell71_after_golden_case_records
    )
):

    cell71_candidate_hash_failures.append(
        "CANDIDATE_HASH_MEMBERSHIP_COUNT_MISMATCH"
    )


if (
    set(
        cell71_candidate_hash_memberships
    )
    !=
    cell71_after_golden_case_ids
):

    cell71_candidate_hash_failures.append(
        "CANDIDATE_HASH_CASE_COVERAGE_MISMATCH"
    )


if (
    len(
        cell71_candidate_hash_memberships
    )
    !=
    len(
        set(
            cell71_candidate_hash_memberships
        )
    )
):

    cell71_candidate_hash_failures.append(
        "CASE_APPEARS_IN_MULTIPLE_CANDIDATE_HASH_BUCKETS"
    )


# ============================================================
# 14) Cross-stratum independent recomputation
# ============================================================

cell71_independent_cross_stratum = {
    candidate_hash:
        case_ids

    for candidate_hash, case_ids
    in cell71_after_candidate_hash_to_case_ids.items()

    if (
        len(
            case_ids
        )
        >
        1
    )
}


cell71_cross_stratum_failure = (
    None

    if (
        cell71_sha(
            cell71_independent_cross_stratum
        )
        ==
        cell71_sha(
            cell71_after_cross_stratum_repeated_candidate_hashes
        )
    )

    else
    "CROSS_STRATUM_RECOMPUTATION_MISMATCH"
)


# ============================================================
# 15) Target-ID validation
# ============================================================

cell71_target_id_count_failures = []


if (
    len(
        cell71_after_target_id_counts
    )
    !=
    572
):

    cell71_target_id_count_failures.append(
        {
            "reason":
                "TARGET_ID_UNIQUE_COUNT_NOT_572",

            "actual":
                len(
                    cell71_after_target_id_counts
                ),
        }
    )


cell71_nonunit_target_counts = {
    target_id:
        count

    for target_id, count
    in cell71_after_target_id_counts.items()

    if (
        count
        !=
        1
    )
}


if (
    cell71_nonunit_target_counts
):

    cell71_target_id_count_failures.append(
        {
            "reason":
                "TARGET_ID_COUNTS_NOT_ALL_ONE",

            "examples":
                list(
                    cell71_nonunit_target_counts.items()
                )[
                    :10
                ],
        }
    )


# ============================================================
# 16) V2 Case / Target admission
# ============================================================

cell71_v2_case_hash_admission_failures = []


cell71_candidate_hash_case_id_set = set(
    cell71_candidate_hash_memberships
)


for case_id in (
    cell65_v2_case_ids
):

    if (
        case_id
        not in
        cell71_candidate_hash_case_id_set
    ):

        cell71_v2_case_hash_admission_failures.append(
            case_id
        )


cell71_v2_target_count_admission_failures = []


for target_id in (
    cell65_v2_target_ids
):

    if (
        cell71_after_target_id_counts.get(
            target_id
        )
        !=
        1
    ):

        cell71_v2_target_count_admission_failures.append(
            {
                "target_evidence_id":
                    target_id,

                "count":
                    cell71_after_target_id_counts.get(
                        target_id
                    ),
            }
        )


# ============================================================
# 17) Packet structural validation
# ============================================================

cell71_packet_structure_failures = []


if (
    cell71_packet_exec_failure
    is not None
):

    cell71_packet_structure_failures.append(
        "PACKET_EXECUTION_FAILED"
    )


if (
    cell71_after_question_drafting_pilot_packets
    is None
):

    cell71_packet_structure_failures.append(
        "AFTER_PACKET_LIST_MISSING"
    )


if (
    cell71_after_question_drafting_pilot_by_id
    is None
):

    cell71_packet_structure_failures.append(
        "AFTER_PACKET_INDEX_MISSING"
    )


if (
    cell71_after_question_drafting_pilot_packets
    is not None

    and

    len(
        cell71_after_question_drafting_pilot_packets
    )
    !=
    23
):

    cell71_packet_structure_failures.append(
        {
            "reason":
                "AFTER_PACKET_COUNT_NOT_23",

            "actual":
                len(
                    cell71_after_question_drafting_pilot_packets
                ),
        }
    )


if (
    cell71_after_question_drafting_pilot_by_id
    is not None

    and

    len(
        cell71_after_question_drafting_pilot_by_id
    )
    !=
    23
):

    cell71_packet_structure_failures.append(
        {
            "reason":
                "AFTER_PACKET_INDEX_COUNT_NOT_23",

            "actual":
                len(
                    cell71_after_question_drafting_pilot_by_id
                ),
        }
    )


# ============================================================
# 18) Packet index independent recomputation
# ============================================================

cell71_packet_index_recompute_failure = None


if (
    cell71_after_question_drafting_pilot_packets
    is not None

    and

    cell71_after_question_drafting_pilot_by_id
    is not None
):

    cell71_independent_packet_by_id = {
        packet[
            "pilot_case_id"
        ]:
            packet

        for packet
        in cell71_after_question_drafting_pilot_packets
    }


    if (
        cell71_sha(
            cell71_independent_packet_by_id
        )
        !=
        cell71_sha(
            cell71_after_question_drafting_pilot_by_id
        )
    ):

        cell71_packet_index_recompute_failure = (
            "PACKET_INDEX_NOT_EXACT_PROJECTION_OF_PACKET_LIST"
        )


# ============================================================
# 19) Legacy leakage across all five AFTER derivatives
# ============================================================

cell71_after_derived_serialized = (
    stable_json_key(
        staged_exercise_v2_after_derived_runtime
    )
)


cell71_legacy_leakage = {
    "case_ids":
        sorted(
            [
                item_id

                for item_id
                in cell65_legacy_exercise_case_ids

                if (
                    item_id
                    in
                    cell71_after_derived_serialized
                )
            ]
        ),

    "task_ids":
        sorted(
            [
                item_id

                for item_id
                in cell65_legacy_exercise_task_ids

                if (
                    item_id
                    in
                    cell71_after_derived_serialized
                )
            ]
        ),

    "target_ids":
        sorted(
            [
                item_id

                for item_id
                in cell65_legacy_exercise_target_ids

                if (
                    item_id
                    in
                    cell71_after_derived_serialized
                )
            ]
        ),
}


# ============================================================
# 20) Resolve BEFORE Exercise packet CORRECTLY
#
# pilot_case_id != golden_case_id
#
# Resolve packet by source_type + stratum, then validate its
# Golden/Task/Target linkage against the active Pilot Selection.
# ============================================================

cell71_before_exercise_selection_rows = [
    row

    for row
    in pilot_selection_rows

    if (
        row[
            "source_type"
        ]
        ==
        CELL71_EXERCISE_SOURCE_TYPE

        and

        row[
            "stratum"
        ]
        ==
        CELL71_EXERCISE_STRATUM
    )
]


cell71_before_exercise_packets = [
    packet

    for packet
    in question_drafting_pilot_packets

    if (
        packet[
            "source_type"
        ]
        ==
        CELL71_EXERCISE_SOURCE_TYPE

        and

        packet[
            "stratum"
        ]
        ==
        CELL71_EXERCISE_STRATUM
    )
]


cell71_before_exercise_resolution_failures = []


if (
    len(
        cell71_before_exercise_selection_rows
    )
    !=
    1
):

    cell71_before_exercise_resolution_failures.append(
        {
            "reason":
                "BEFORE_EXERCISE_SELECTION_COUNT_NOT_1",

            "actual":
                len(
                    cell71_before_exercise_selection_rows
                ),
        }
    )


if (
    len(
        cell71_before_exercise_packets
    )
    !=
    1
):

    cell71_before_exercise_resolution_failures.append(
        {
            "reason":
                "BEFORE_EXERCISE_PACKET_COUNT_NOT_1",

            "actual":
                len(
                    cell71_before_exercise_packets
                ),
        }
    )


cell71_before_exercise_selection = (
    cell71_before_exercise_selection_rows[
        0
    ]

    if (
        len(
            cell71_before_exercise_selection_rows
        )
        ==
        1
    )

    else None
)


cell71_before_exercise_packet = (
    cell71_before_exercise_packets[
        0
    ]

    if (
        len(
            cell71_before_exercise_packets
        )
        ==
        1
    )

    else None
)


if (
    cell71_before_exercise_selection
    is not None

    and

    cell71_before_exercise_packet
    is not None
):

    for field_name in (
        [
            "golden_case_id",
            "task_draft_id",
            "target_evidence_id",
        ]
    ):

        if (
            cell71_before_exercise_packet[
                field_name
            ]
            !=
            cell71_before_exercise_selection[
                field_name
            ]
        ):

            cell71_before_exercise_resolution_failures.append(
                {
                    "reason":
                        "BEFORE_PACKET_SELECTION_LINKAGE_MISMATCH",

                    "field":
                        field_name,

                    "packet":
                        cell71_before_exercise_packet[
                            field_name
                        ],

                    "selection":
                        cell71_before_exercise_selection[
                            field_name
                        ],
                }
            )


cell71_legacy_exercise_packet_id = (
    cell71_before_exercise_packet[
        "pilot_case_id"
    ]

    if (
        cell71_before_exercise_packet
        is not None
    )

    else None
)


# ============================================================
# 21) Resolve AFTER Exercise packet CORRECTLY
# ============================================================

cell71_after_exercise_selection_rows = [
    row

    for row
    in cell71_after_pilot_selection_rows

    if (
        row[
            "source_type"
        ]
        ==
        CELL71_EXERCISE_SOURCE_TYPE

        and

        row[
            "stratum"
        ]
        ==
        CELL71_EXERCISE_STRATUM
    )
]


cell71_after_exercise_packets = [
    packet

    for packet
    in (
        cell71_after_question_drafting_pilot_packets
        or
        []
    )

    if (
        packet[
            "source_type"
        ]
        ==
        CELL71_EXERCISE_SOURCE_TYPE

        and

        packet[
            "stratum"
        ]
        ==
        CELL71_EXERCISE_STRATUM
    )
]


cell71_after_exercise_resolution_failures = []


if (
    len(
        cell71_after_exercise_selection_rows
    )
    !=
    1
):

    cell71_after_exercise_resolution_failures.append(
        {
            "reason":
                "AFTER_EXERCISE_SELECTION_COUNT_NOT_1",

            "actual":
                len(
                    cell71_after_exercise_selection_rows
                ),
        }
    )


if (
    len(
        cell71_after_exercise_packets
    )
    !=
    1
):

    cell71_after_exercise_resolution_failures.append(
        {
            "reason":
                "AFTER_EXERCISE_PACKET_COUNT_NOT_1",

            "actual":
                len(
                    cell71_after_exercise_packets
                ),
        }
    )


cell71_after_exercise_selection = (
    cell71_after_exercise_selection_rows[
        0
    ]

    if (
        len(
            cell71_after_exercise_selection_rows
        )
        ==
        1
    )

    else None
)


cell71_v2_exercise_packet = (
    cell71_after_exercise_packets[
        0
    ]

    if (
        len(
            cell71_after_exercise_packets
        )
        ==
        1
    )

    else None
)


if (
    cell71_after_exercise_selection
    is not None

    and

    cell71_v2_exercise_packet
    is not None
):

    for field_name in (
        [
            "golden_case_id",
            "task_draft_id",
            "target_evidence_id",
        ]
    ):

        if (
            cell71_v2_exercise_packet[
                field_name
            ]
            !=
            cell71_after_exercise_selection[
                field_name
            ]
        ):

            cell71_after_exercise_resolution_failures.append(
                {
                    "reason":
                        "AFTER_PACKET_SELECTION_LINKAGE_MISMATCH",

                    "field":
                        field_name,

                    "packet":
                        cell71_v2_exercise_packet[
                            field_name
                        ],

                    "selection":
                        cell71_after_exercise_selection[
                            field_name
                        ],
                }
            )


# The staged Pilot selection itself is the authoritative
# expected Case/Task/Target chain.
if (
    cell71_v2_exercise_packet
    is not None
):

    for field_name in (
        [
            "golden_case_id",
            "task_draft_id",
            "target_evidence_id",
        ]
    ):

        if (
            cell71_v2_exercise_packet[
                field_name
            ]
            !=
            staged_exercise_v2_pilot_replacement_row[
                field_name
            ]
        ):

            cell71_after_exercise_resolution_failures.append(
                {
                    "reason":
                        "AFTER_PACKET_STAGED_V2_LINKAGE_MISMATCH",

                    "field":
                        field_name,

                    "packet":
                        cell71_v2_exercise_packet[
                            field_name
                        ],

                    "expected":
                        staged_exercise_v2_pilot_replacement_row[
                            field_name
                        ],
                }
            )


cell71_v2_exercise_packet_id = (
    cell71_v2_exercise_packet[
        "pilot_case_id"
    ]

    if (
        cell71_v2_exercise_packet
        is not None
    )

    else None
)


# ============================================================
# 22) Preserve exact 22 NON-Exercise packets
#
# Filter by semantic stratum identity, NOT by pilot_case_id.
# ============================================================

cell71_before_nonexercise_packets = [
    packet

    for packet
    in question_drafting_pilot_packets

    if not (
        packet[
            "source_type"
        ]
        ==
        CELL71_EXERCISE_SOURCE_TYPE

        and

        packet[
            "stratum"
        ]
        ==
        CELL71_EXERCISE_STRATUM
    )
]


cell71_after_nonexercise_packets = [
    packet

    for packet
    in (
        cell71_after_question_drafting_pilot_packets
        or
        []
    )

    if not (
        packet[
            "source_type"
        ]
        ==
        CELL71_EXERCISE_SOURCE_TYPE

        and

        packet[
            "stratum"
        ]
        ==
        CELL71_EXERCISE_STRATUM
    )
]


cell71_packet_preservation_failures = []


if (
    len(
        cell71_before_nonexercise_packets
    )
    !=
    22
):

    cell71_packet_preservation_failures.append(
        {
            "reason":
                "BEFORE_NONEXERCISE_PACKET_COUNT_NOT_22",

            "actual":
                len(
                    cell71_before_nonexercise_packets
                ),
        }
    )


if (
    len(
        cell71_after_nonexercise_packets
    )
    !=
    22
):

    cell71_packet_preservation_failures.append(
        {
            "reason":
                "AFTER_NONEXERCISE_PACKET_COUNT_NOT_22",

            "actual":
                len(
                    cell71_after_nonexercise_packets
                ),
        }
    )


if (
    cell71_identity_keyed_sha(
        cell71_before_nonexercise_packets,
        "pilot_case_id",
    )
    !=
    cell71_identity_keyed_sha(
        cell71_after_nonexercise_packets,
        "pilot_case_id",
    )
):

    cell71_packet_preservation_failures.append(
        "NONEXERCISE_PACKET_CONTENT_CHANGED"
    )


# ============================================================
# 23) Explicit V2 Exercise packet semantic validation
# ============================================================

cell71_v2_exercise_packet_failures = []


if (
    cell71_v2_exercise_packet
    is None
):

    cell71_v2_exercise_packet_failures.append(
        "V2_EXERCISE_PACKET_MISSING"
    )


else:

    expected_fields = {
        "source_type":
            CELL71_EXERCISE_SOURCE_TYPE,

        "stratum":
            CELL71_EXERCISE_STRATUM,

        "golden_case_id":
            staged_exercise_v2_pilot_replacement_row[
                "golden_case_id"
            ],

        "task_draft_id":
            staged_exercise_v2_pilot_replacement_row[
                "task_draft_id"
            ],

        "target_evidence_id":
            staged_exercise_v2_pilot_replacement_row[
                "target_evidence_id"
            ],
    }


    for field_name, expected_value in (
        expected_fields.items()
    ):

        actual_value = (
            cell71_v2_exercise_packet.get(
                field_name
            )
        )


        if (
            actual_value
            !=
            expected_value
        ):

            cell71_v2_exercise_packet_failures.append(
                {
                    "field":
                        field_name,

                    "expected":
                        expected_value,

                    "actual":
                        actual_value,
                }
            )


# ============================================================
# 24) Old Exercise packet identity must disappear
# ============================================================

cell71_old_exercise_packet_removal_failure = None


if (
    cell71_legacy_exercise_packet_id
    is not None

    and

    cell71_after_question_drafting_pilot_by_id
    is not None

    and

    cell71_legacy_exercise_packet_id
    in
    cell71_after_question_drafting_pilot_by_id
):

    cell71_old_exercise_packet_removal_failure = (
        "LEGACY_EXERCISE_PILOT_PACKET_ID_SURVIVES_AFTER_REBUILD"
    )


# ============================================================
# 25) Exactly one packet identity removed + one added
# ============================================================

cell71_before_packet_ids = {
    packet[
        "pilot_case_id"
    ]

    for packet
    in question_drafting_pilot_packets
}


cell71_after_packet_ids = {
    packet[
        "pilot_case_id"
    ]

    for packet
    in (
        cell71_after_question_drafting_pilot_packets
        or
        []
    )
}


cell71_removed_packet_ids = sorted(
    cell71_before_packet_ids
    -
    cell71_after_packet_ids
)


cell71_added_packet_ids = sorted(
    cell71_after_packet_ids
    -
    cell71_before_packet_ids
)


cell71_packet_identity_transition_failure = None


if not (
    cell71_legacy_exercise_packet_id
    is not None

    and

    cell71_v2_exercise_packet_id
    is not None

    and

    cell71_removed_packet_ids
    ==
    [
        cell71_legacy_exercise_packet_id
    ]

    and

    cell71_added_packet_ids
    ==
    [
        cell71_v2_exercise_packet_id
    ]
):

    cell71_packet_identity_transition_failure = {
        "legacy_exercise_packet_id":
            cell71_legacy_exercise_packet_id,

        "v2_exercise_packet_id":
            cell71_v2_exercise_packet_id,

        "removed":
            cell71_removed_packet_ids,

        "added":
            cell71_added_packet_ids,
    }


# ============================================================
# 26) Verify V2 pilot_case_id itself is internally consistent
#
# If the original builder function is available, independently
# recompute from the packet's own source-proven identity inputs.
# ============================================================

cell71_v2_pilot_case_id_recompute_failure = None


if (
    cell71_v2_exercise_packet
    is not None

    and

    "build_pilot_case_id"
    in
    globals()
):

    try:

        cell71_recomputed_v2_pilot_case_id = (
            build_pilot_case_id(
                cell71_v2_exercise_packet[
                    "golden_case_id"
                ],
                cell71_v2_exercise_packet[
                    "task_draft_id"
                ],
                cell71_v2_exercise_packet[
                    "target_evidence_id"
                ],
            )
        )


        if (
            cell71_recomputed_v2_pilot_case_id
            !=
            cell71_v2_exercise_packet[
                "pilot_case_id"
            ]
        ):

            cell71_v2_pilot_case_id_recompute_failure = {
                "expected":
                    cell71_recomputed_v2_pilot_case_id,

                "actual":
                    cell71_v2_exercise_packet[
                        "pilot_case_id"
                    ],
            }


    except Exception as exc:

        # Signature may differ; do not invent a recomputation
        # contract. The exact Cell57 builder remains primary
        # evidence in that case.
        cell71_recomputed_v2_pilot_case_id = None


else:

    cell71_recomputed_v2_pilot_case_id = None


# ============================================================
# 27) Historical objects intentionally remain untouched
# ============================================================

cell71_historical_rebuild_execution_status = {
    variable_name:
        "PRESERVED_UNCHANGED_NOT_REBUILT"

    for variable_name
    in sorted(
        CELL71_HISTORICAL_NOT_REBUILT
    )
}


# ============================================================
# 28) Full AFTER transaction clone
# ============================================================

cell71_full_after_transaction_clone = {
    "canonical_registries":
        cell71_clone(
            cell67_after_state
        ),

    "canonical_indexes":
        cell71_clone(
            cell67_after_indexes
        ),

    "rebuilt_current_state_derivatives":
        cell71_clone(
            staged_exercise_v2_after_derived_runtime
        ),

    "historical_preserve_policy":
        {
            variable_name:
                "PRESERVE_EXISTING_RUNTIME_OBJECT"

            for variable_name
            in sorted(
                cell69_v2_historical_preserve_objects
            )
        },

    "safe_no_action_policy":
        {
            variable_name:
                "NO_TRANSACTION_ACTION"

            for variable_name
            in sorted(
                cell69_v2_safe_no_action_objects
            )
        },
}


cell71_full_after_transaction_sha256 = (
    cell71_sha(
        cell71_full_after_transaction_clone
    )
)


# ============================================================
# 29) Aggregate failures
# ============================================================

cell71_after_rebuild_failures = []


if (
    cell71_packet_exec_failure
    is not None
):

    cell71_after_rebuild_failures.append(
        {
            "reason":
                "PACKET_EXEC_FAILURE",

            "detail":
                cell71_packet_exec_failure,
        }
    )


if (
    cell71_candidate_hash_failures
):

    cell71_after_rebuild_failures.append(
        {
            "reason":
                "CANDIDATE_HASH_FAILURE",

            "detail":
                cell71_candidate_hash_failures,
        }
    )


if (
    cell71_cross_stratum_failure
    is not None
):

    cell71_after_rebuild_failures.append(
        cell71_cross_stratum_failure
    )


if (
    cell71_target_id_count_failures
):

    cell71_after_rebuild_failures.append(
        {
            "reason":
                "TARGET_ID_COUNT_FAILURE",

            "detail":
                cell71_target_id_count_failures,
        }
    )


if (
    cell71_v2_case_hash_admission_failures
):

    cell71_after_rebuild_failures.append(
        {
            "reason":
                "V2_CASE_HASH_ADMISSION_FAILURE",

            "count":
                len(
                    cell71_v2_case_hash_admission_failures
                ),
        }
    )


if (
    cell71_v2_target_count_admission_failures
):

    cell71_after_rebuild_failures.append(
        {
            "reason":
                "V2_TARGET_COUNT_ADMISSION_FAILURE",

            "count":
                len(
                    cell71_v2_target_count_admission_failures
                ),
        }
    )


if (
    cell71_packet_structure_failures
):

    cell71_after_rebuild_failures.append(
        {
            "reason":
                "PACKET_STRUCTURE_FAILURE",

            "detail":
                cell71_packet_structure_failures,
        }
    )


if (
    cell71_packet_index_recompute_failure
    is not None
):

    cell71_after_rebuild_failures.append(
        cell71_packet_index_recompute_failure
    )


if (
    any(
        cell71_legacy_leakage.values()
    )
):

    cell71_after_rebuild_failures.append(
        {
            "reason":
                "LEGACY_ID_LEAKAGE",

            "detail":
                cell71_legacy_leakage,
        }
    )


if (
    cell71_before_exercise_resolution_failures
):

    cell71_after_rebuild_failures.append(
        {
            "reason":
                "BEFORE_EXERCISE_PACKET_RESOLUTION_FAILURE",

            "detail":
                cell71_before_exercise_resolution_failures,
        }
    )


if (
    cell71_after_exercise_resolution_failures
):

    cell71_after_rebuild_failures.append(
        {
            "reason":
                "AFTER_EXERCISE_PACKET_RESOLUTION_FAILURE",

            "detail":
                cell71_after_exercise_resolution_failures,
        }
    )


if (
    cell71_packet_preservation_failures
):

    cell71_after_rebuild_failures.append(
        {
            "reason":
                "NONEXERCISE_PACKET_PRESERVATION_FAILURE",

            "detail":
                cell71_packet_preservation_failures,
        }
    )


if (
    cell71_v2_exercise_packet_failures
):

    cell71_after_rebuild_failures.append(
        {
            "reason":
                "V2_EXERCISE_PACKET_FAILURE",

            "detail":
                cell71_v2_exercise_packet_failures,
        }
    )


if (
    cell71_old_exercise_packet_removal_failure
    is not None
):

    cell71_after_rebuild_failures.append(
        cell71_old_exercise_packet_removal_failure
    )


if (
    cell71_packet_identity_transition_failure
    is not None
):

    cell71_after_rebuild_failures.append(
        {
            "reason":
                "PACKET_IDENTITY_TRANSITION_FAILURE",

            "detail":
                cell71_packet_identity_transition_failure,
        }
    )


if (
    cell71_v2_pilot_case_id_recompute_failure
    is not None
):

    cell71_after_rebuild_failures.append(
        {
            "reason":
                "V2_PILOT_CASE_ID_RECOMPUTE_FAILURE",

            "detail":
                cell71_v2_pilot_case_id_recompute_failure,
        }
    )


# ============================================================
# 30) Freeze AFTER-derived rebuild contract
# ============================================================

EXERCISE_V2_TRUE_DERIVED_AFTER_REBUILD_V1_DRAFT = {
    "version":
        "exercise_v2_true_derived_after_rebuild_v1_draft",

    "status":
        (
            "AFTER_CLONE_5_DERIVED_REBUILDS_VALIDATED"

            if (
                len(
                    cell71_prerequisite_failures
                )
                ==
                0

                and

                len(
                    cell71_after_rebuild_failures
                )
                ==
                0
            )

            else
            "AFTER_DERIVED_REBUILD_VALIDATION_FAILED"
        ),

    "rebuilt_objects":
        sorted(
            CELL71_TRUE_DERIVED_OBJECTS
        ),

    "rebuilt_object_fingerprints":
        cell71_clone(
            cell71_after_derived_fingerprints
        ),

    "historical_objects_not_rebuilt":
        sorted(
            CELL71_HISTORICAL_NOT_REBUILT
        ),

    "exercise_pilot_transition": {
        "legacy_golden_case_id":
            (
                cell71_before_exercise_selection[
                    "golden_case_id"
                ]
                if (
                    cell71_before_exercise_selection
                    is not None
                )
                else None
            ),

        "v2_golden_case_id":
            staged_exercise_v2_pilot_replacement_row[
                "golden_case_id"
            ],

        "legacy_pilot_case_id":
            cell71_legacy_exercise_packet_id,

        "v2_pilot_case_id":
            cell71_v2_exercise_packet_id,

        "removed_packet_ids":
            cell71_removed_packet_ids,

        "added_packet_ids":
            cell71_added_packet_ids,
    },

    "legacy_leakage":
        cell71_clone(
            cell71_legacy_leakage
        ),

    "full_after_transaction_sha256":
        cell71_full_after_transaction_sha256,

    "actual_activation_status":
        "NOT_STARTED",

    "rollback_status":
        "NOT_NEEDED",

    "question_generation_status":
        "NOT_STARTED",

    "expected_answer_generation_status":
        "NOT_STARTED",

    "artifact_write_status":
        "NONE",
}


# ============================================================
# 31) Full transaction-clone readiness
# ============================================================

cell71_full_transaction_clone_ready = (
    EXERCISE_V2_TRUE_DERIVED_AFTER_REBUILD_V1_DRAFT[
        "status"
    ]
    ==
    "AFTER_CLONE_5_DERIVED_REBUILDS_VALIDATED"

    and

    cell67_transaction_clone_ready
)


cell71_transaction_status = (
    "FULL_AFTER_TRANSACTION_CLONE_VALIDATED_"
    "ACTIVATION_NOT_EXECUTED"

    if (
        cell71_full_transaction_clone_ready
    )

    else
    "FULL_AFTER_TRANSACTION_CLONE_NOT_READY"
)


# ============================================================
# 32) No-mutation recheck
# ============================================================

cell71_live_active_snapshot_after = {
    "golden":
        cell71_sha(
            golden_case_records
        ),

    "auto":
        cell71_sha(
            auto_case_records
        ),

    "groundability":
        cell71_sha(
            case_groundability_precheck
        ),

    "task_design":
        cell71_sha(
            case_task_design
        ),

    "tasks":
        cell71_sha(
            task_draft_plan_rows
        ),

    "targets":
        cell71_sha(
            target_evidence_records
        ),

    "pilots":
        cell71_sha(
            pilot_selection_rows
        ),
}


cell71_live_true_derived_snapshot_after = {
    variable_name:
        cell71_sha(
            globals()[
                variable_name
            ]
        )

    for variable_name
    in CELL71_TRUE_DERIVED_OBJECTS
}


cell71_live_historical_snapshot_after = {
    variable_name:
        cell71_sha(
            globals()[
                variable_name
            ]
        )

    for variable_name
    in CELL71_HISTORICAL_NOT_REBUILT
}


cell71_transaction_before_sha_after = (
    cell71_sha(
        cell67_active_state_before
    )
)


cell71_transaction_after_sha_after = (
    cell71_sha(
        cell67_after_state
    )
)


cell71_transaction_after_indexes_sha_after = (
    cell71_sha(
        cell67_after_indexes
    )
)


# ============================================================
# 33) Gates
# ============================================================

cell71_gates = {
    "cell70_exact_before_reproduction_and_cell69_reconciliation_are_valid":
        (
            len(
                cell71_prerequisite_failures
            )
            ==
            0
        ),

    "exact_proven_cell57_packet_builder_executes_on_after_clone":
        (
            cell71_packet_exec_failure
            is None
        ),

    "candidate_hash_index_covers_each_of_724_after_cases_exactly_once":
        (
            len(
                cell71_candidate_hash_failures
            )
            ==
            0

            and

            len(
                cell71_candidate_hash_memberships
            )
            ==
            724
        ),

    "cross_stratum_repeated_hashes_recompute_exactly":
        (
            cell71_cross_stratum_failure
            is None
        ),

    "target_id_counts_contains_exactly_572_unique_targets_all_with_count_one":
        (
            len(
                cell71_target_id_count_failures
            )
            ==
            0

            and

            len(
                cell71_after_target_id_counts
            )
            ==
            572
        ),

    "all_38_v2_case_ids_are_admitted_to_after_candidate_hash_index":
        (
            len(
                cell71_v2_case_hash_admission_failures
            )
            ==
            0
        ),

    "all_38_v2_target_ids_are_admitted_exactly_once":
        (
            len(
                cell71_v2_target_count_admission_failures
            )
            ==
            0
        ),

    "after_question_drafting_packets_and_index_both_contain_exactly_23_records":
        (
            len(
                cell71_packet_structure_failures
            )
            ==
            0

            and

            cell71_after_question_drafting_pilot_packets
            is not None

            and

            cell71_after_question_drafting_pilot_by_id
            is not None

            and

            len(
                cell71_after_question_drafting_pilot_packets
            )
            ==
            23

            and

            len(
                cell71_after_question_drafting_pilot_by_id
            )
            ==
            23
        ),

    "after_packet_index_is_exact_projection_of_after_packet_list":
        (
            cell71_packet_index_recompute_failure
            is None
        ),

    "no_legacy_exercise_case_task_or_target_id_survives_in_five_rebuilt_objects":
        (
            not any(
                cell71_legacy_leakage.values()
            )
        ),

    "legacy_exercise_packet_is_resolved_by_actual_source_and_stratum":
        (
            len(
                cell71_before_exercise_resolution_failures
            )
            ==
            0

            and

            cell71_legacy_exercise_packet_id
            is not None
        ),

    "v2_exercise_packet_is_resolved_by_actual_source_and_stratum":
        (
            len(
                cell71_after_exercise_resolution_failures
            )
            ==
            0

            and

            cell71_v2_exercise_packet_id
            is not None
        ),

    "all_22_nonexercise_question_drafting_packets_are_preserved_exactly":
        (
            len(
                cell71_packet_preservation_failures
            )
            ==
            0
        ),

    "v2_exercise_packet_links_to_exact_staged_case_task_target":
        (
            len(
                cell71_v2_exercise_packet_failures
            )
            ==
            0
        ),

    "legacy_exercise_question_drafting_packet_identity_is_removed":
        (
            cell71_old_exercise_packet_removal_failure
            is None
        ),

    "packet_identity_transition_removes_exactly_one_old_question_pilot_and_adds_exactly_one_new_question_pilot":
        (
            cell71_packet_identity_transition_failure
            is None
        ),

    "all_three_reclassified_historical_runtime_objects_remain_unrebuilt_and_byte_stable":
        (
            cell71_live_historical_snapshot_before
            ==
            cell71_live_historical_snapshot_after

            and

            all(
                status
                ==
                "PRESERVED_UNCHANGED_NOT_REBUILT"

                for status
                in cell71_historical_rebuild_execution_status.values()
            )
        ),

    "after_derived_rebuild_contract_is_validated":
        (
            EXERCISE_V2_TRUE_DERIVED_AFTER_REBUILD_V1_DRAFT[
                "status"
            ]
            ==
            "AFTER_CLONE_5_DERIVED_REBUILDS_VALIDATED"

            and

            len(
                cell71_after_rebuild_failures
            )
            ==
            0
        ),

    "full_after_transaction_clone_is_ready_but_not_activated":
        (
            cell71_full_transaction_clone_ready
            is True

            and

            cell71_transaction_status
            ==
            "FULL_AFTER_TRANSACTION_CLONE_VALIDATED_"
            "ACTIVATION_NOT_EXECUTED"
        ),

    "all_live_canonical_registries_remain_byte_stable":
        (
            cell71_live_active_snapshot_before
            ==
            cell71_live_active_snapshot_after
        ),

    "all_five_live_current_state_derived_objects_remain_byte_stable":
        (
            cell71_live_true_derived_snapshot_before
            ==
            cell71_live_true_derived_snapshot_after
        ),

    "cell67_before_after_and_index_clones_remain_byte_stable":
        (
            cell71_transaction_before_sha_before
            ==
            cell71_transaction_before_sha_after

            and

            cell71_transaction_after_sha_before
            ==
            cell71_transaction_after_sha_after

            and

            cell71_transaction_after_indexes_sha_before
            ==
            cell71_transaction_after_indexes_sha_after
        ),

    "actual_activation_not_started":
        (
            EXERCISE_V2_TRUE_DERIVED_AFTER_REBUILD_V1_DRAFT[
                "actual_activation_status"
            ]
            ==
            "NOT_STARTED"
        ),

    "question_generation_not_started":
        (
            EXERCISE_V2_TRUE_DERIVED_AFTER_REBUILD_V1_DRAFT[
                "question_generation_status"
            ]
            ==
            "NOT_STARTED"
        ),

    "expected_answer_generation_not_started":
        (
            EXERCISE_V2_TRUE_DERIVED_AFTER_REBUILD_V1_DRAFT[
                "expected_answer_generation_status"
            ]
            ==
            "NOT_STARTED"
        ),

    "no_artifact_write":
        (
            EXERCISE_V2_TRUE_DERIVED_AFTER_REBUILD_V1_DRAFT[
                "artifact_write_status"
            ]
            ==
            "NONE"
        ),
}


cell_71_pass = all(
    cell71_gates.values()
)


# ============================================================
# 34) Output
# ============================================================

print(
    "=== EXERCISE V2 TRUE-DERIVED AFTER REBUILD — REPAIRED ==="
)


print(
    "\n=== PREREQUISITES ==="
)


print(
    "Failures:",
    cell71_prerequisite_failures
)


print(
    "\n=== FIVE REBUILT AFTER OBJECTS ==="
)


for variable_name in sorted(
    CELL71_TRUE_DERIVED_OBJECTS
):

    value = (
        staged_exercise_v2_after_derived_runtime[
            variable_name
        ]
    )


    print(
        "\nOBJECT:",
        variable_name
    )


    print(
        "Type:",
        (
            type(
                value
            ).__name__
            if value is not None
            else None
        )
    )


    print(
        "Length:",
        (
            len(
                value
            )
            if value is not None
            else None
        )
    )


    print(
        "SHA256:",
        cell71_after_derived_fingerprints[
            variable_name
        ]
    )


print(
    "\nPacket execution failure:",
    cell71_packet_exec_failure
)


print(
    "\n=== CANDIDATE HASH ==="
)


print(
    "Buckets:",
    len(
        cell71_after_candidate_hash_to_case_ids
    )
)


print(
    "Case memberships:",
    len(
        cell71_candidate_hash_memberships
    )
)


print(
    "Failures:",
    cell71_candidate_hash_failures
)


print(
    "Cross-stratum repeated hashes:",
    len(
        cell71_after_cross_stratum_repeated_candidate_hashes
    )
)


print(
    "Cross-stratum failure:",
    cell71_cross_stratum_failure
)


print(
    "\n=== TARGET IDS ==="
)


print(
    "Unique IDs:",
    len(
        cell71_after_target_id_counts
    )
)


print(
    "Non-unit counts:",
    len(
        cell71_nonunit_target_counts
    )
)


print(
    "Failures:",
    cell71_target_id_count_failures
)


print(
    "\n=== V2 ADMISSION ==="
)


print(
    "Missing V2 Cases:",
    len(
        cell71_v2_case_hash_admission_failures
    )
)


print(
    "Missing/non-unit V2 Targets:",
    len(
        cell71_v2_target_count_admission_failures
    )
)


print(
    "\n=== QUESTION DRAFTING PACKETS ==="
)


print(
    "AFTER packets:",
    (
        len(
            cell71_after_question_drafting_pilot_packets
        )
        if (
            cell71_after_question_drafting_pilot_packets
            is not None
        )
        else None
    )
)


print(
    "AFTER packet index:",
    (
        len(
            cell71_after_question_drafting_pilot_by_id
        )
        if (
            cell71_after_question_drafting_pilot_by_id
            is not None
        )
        else None
    )
)


print(
    "Structure failures:",
    cell71_packet_structure_failures
)


print(
    "Index recompute failure:",
    cell71_packet_index_recompute_failure
)


print(
    "\n=== EXERCISE PACKET RESOLUTION ==="
)


print(
    "Legacy Golden Case:",
    (
        cell71_before_exercise_selection[
            "golden_case_id"
        ]
        if (
            cell71_before_exercise_selection
            is not None
        )
        else None
    )
)


print(
    "Legacy question-pilot ID:",
    cell71_legacy_exercise_packet_id
)


print(
    "V2 Golden Case:",
    staged_exercise_v2_pilot_replacement_row[
        "golden_case_id"
    ]
)


print(
    "V2 question-pilot ID:",
    cell71_v2_exercise_packet_id
)


print(
    "Before resolution failures:",
    cell71_before_exercise_resolution_failures
)


print(
    "After resolution failures:",
    cell71_after_exercise_resolution_failures
)


print(
    "V2 packet linkage failures:",
    cell71_v2_exercise_packet_failures
)


print(
    "\nNon-Exercise packet preservation failures:",
    cell71_packet_preservation_failures
)


print(
    "Removed packet IDs:",
    cell71_removed_packet_ids
)


print(
    "Added packet IDs:",
    cell71_added_packet_ids
)


print(
    "Old packet removal failure:",
    cell71_old_exercise_packet_removal_failure
)


print(
    "Packet identity transition failure:",
    cell71_packet_identity_transition_failure
)


print(
    "Independent V2 pilot-case ID recompute:",
    cell71_recomputed_v2_pilot_case_id
)


print(
    "Pilot-case ID recompute failure:",
    cell71_v2_pilot_case_id_recompute_failure
)


print(
    "\n=== LEGACY LEAKAGE ==="
)


print(
    cell71_legacy_leakage
)


print(
    "\n=== HISTORICAL OBJECT POLICY ==="
)


for variable_name, status in sorted(
    cell71_historical_rebuild_execution_status.items()
):

    print(
        "-",
        variable_name,
        ":",
        status
    )


print(
    "\nAfter Derived Bundle SHA256:",
    cell71_after_derived_bundle_sha256
)


print(
    "Full AFTER Transaction SHA256:",
    cell71_full_after_transaction_sha256
)


print(
    "\nAfter rebuild failures:",
    cell71_after_rebuild_failures
)


print(
    "\nTransaction status:",
    cell71_transaction_status
)


print(
    "\n=== ACTIVE COUNTS — MUST REMAIN UNCHANGED ==="
)


print(
    "Golden:",
    len(
        golden_case_records
    )
)


print(
    "Auto:",
    len(
        cell65_active_auto_rows
    )
)


print(
    "Groundability:",
    len(
        cell65_active_groundability_rows
    )
)


print(
    "Task Design:",
    len(
        cell65_active_task_design_rows
    )
)


print(
    "Tasks:",
    len(
        task_draft_plan_rows
    )
)


print(
    "Targets:",
    len(
        target_evidence_records
    )
)


print(
    "Pilots:",
    len(
        pilot_selection_rows
    )
)


print(
    "\n=== CODE CELL 71 GATES ==="
)


for gate_name, passed in (
    cell71_gates.items()
):

    print(
        gate_name,
        ":",
        "PASS"
        if passed
        else "FAIL"
    )


print(
    "\nFAILED GATES:",
    [
        gate_name

        for gate_name, passed
        in cell71_gates.items()

        if not passed
    ]
)


print(
    "\nCODE CELL 71 GATE:",
    "PASS"
    if cell_71_pass
    else "FAIL"
)


assert cell_71_pass, (
    "The five true derived runtime objects still do not "
    "validate on the AFTER clone. Do not activate the Exercise "
    "V2 transaction. Use the explicit Exercise packet "
    "resolution, linkage, leakage and identity-transition "
    "diagnostics from this SAME Code Cell 71."
)
```

```python
# Code Cell 73 — REPAIRED

# ============================================================
# Final Semantic Inspection of the V2 Exercise Pilot
#
# Repair:
#   The previous Cell73 mixed two different concerns:
#
#   1) Is the V2 pilot packet structurally valid / inspectable?
#   2) Does every semantic field equal a hard-coded string?
#
# The first is a machine gate.
# The second must be inspected from the actual staged V2
# Case / Task / Target contracts and evidence.
#
# This Cell therefore:
#
#   - resolves the exact V2 Case / Task / Target / Pilot;
#   - validates their linkage;
#   - validates multi-part textual evidence + locators;
#   - validates zero Legacy leakage;
#   - prints semantic contract values as diagnostics;
#   - prints the FULL actual target evidence;
#
# NO activation.
# NO mutation.
# NO question generation.
# NO artifact write.
# ============================================================


# ============================================================
# 1) Authorization
# ============================================================

CELL73_EXPLICIT_ACTIVATION_AUTHORIZATION = (
    "PROVIDED_BY_USER_IN_CHAT"
)


# ============================================================
# 2) Helpers
# ============================================================

def cell73_sha(
    value,
):

    return hashlib.sha256(
        stable_json_key(
            value
        ).encode(
            "utf-8"
        )
    ).hexdigest()


def cell73_clone(
    value,
):

    return cell52_safe_clone(
        value
    )


# ============================================================
# 3) Snapshot live state
# ============================================================

cell73_active_snapshot_before = {
    "golden":
        cell73_sha(
            golden_case_records
        ),

    "auto":
        cell73_sha(
            auto_case_records
        ),

    "groundability":
        cell73_sha(
            case_groundability_precheck
        ),

    "task_design":
        cell73_sha(
            case_task_design
        ),

    "tasks":
        cell73_sha(
            task_draft_plan_rows
        ),

    "targets":
        cell73_sha(
            target_evidence_records
        ),

    "pilots":
        cell73_sha(
            pilot_selection_rows
        ),
}


# ============================================================
# 4) Preconditions
# ============================================================

cell73_prerequisite_failures = []


if not (
    cell_72_pass
):

    cell73_prerequisite_failures.append(
        "CELL72_NOT_PASS"
    )


if not (
    cell72_technical_activation_ready
):

    cell73_prerequisite_failures.append(
        "CELL72_TECHNICAL_ACTIVATION_NOT_READY"
    )


if (
    CELL73_EXPLICIT_ACTIVATION_AUTHORIZATION
    !=
    "PROVIDED_BY_USER_IN_CHAT"
):

    cell73_prerequisite_failures.append(
        "EXPLICIT_AUTHORIZATION_NOT_RECORDED"
    )


if not (
    cell_71_pass
):

    cell73_prerequisite_failures.append(
        "CELL71_NOT_PASS"
    )


# ============================================================
# 5) Authoritative expected V2 pilot chain
# ============================================================

cell73_expected_chain = {
    "golden_case_id":
        staged_exercise_v2_pilot_replacement_row[
            "golden_case_id"
        ],

    "task_draft_id":
        staged_exercise_v2_pilot_replacement_row[
            "task_draft_id"
        ],

    "target_evidence_id":
        staged_exercise_v2_pilot_replacement_row[
            "target_evidence_id"
        ],
}


# ============================================================
# 6) Resolve exact V2 Pilot packet
# ============================================================

cell73_v2_packet = (
    cell73_clone(
        cell71_v2_exercise_packet
    )
    if (
        cell71_v2_exercise_packet
        is not None
    )
    else None
)


# ============================================================
# 7) Resolve authoritative AFTER Case
# ============================================================

cell73_case_matches = [
    row

    for row
    in cell67_after_state[
        "golden_case_records"
    ]

    if (
        row[
            "golden_case_id"
        ]
        ==
        cell73_expected_chain[
            "golden_case_id"
        ]
    )
]


cell73_v2_case = (
    cell73_clone(
        cell73_case_matches[
            0
        ]
    )

    if (
        len(
            cell73_case_matches
        )
        ==
        1
    )

    else None
)


# ============================================================
# 8) Resolve authoritative AFTER Task
# ============================================================

cell73_task_matches = [
    row

    for row
    in cell67_after_state[
        "task_draft_plan_rows"
    ]

    if (
        row[
            "task_draft_id"
        ]
        ==
        cell73_expected_chain[
            "task_draft_id"
        ]
    )
]


cell73_v2_task = (
    cell73_clone(
        cell73_task_matches[
            0
        ]
    )

    if (
        len(
            cell73_task_matches
        )
        ==
        1
    )

    else None
)


# ============================================================
# 9) Resolve authoritative AFTER Target
# ============================================================

cell73_target_matches = [
    row

    for row
    in cell67_after_state[
        "target_evidence_records"
    ]

    if (
        row[
            "target_evidence_id"
        ]
        ==
        cell73_expected_chain[
            "target_evidence_id"
        ]
    )
]


cell73_v2_target = (
    cell73_clone(
        cell73_target_matches[
            0
        ]
    )

    if (
        len(
            cell73_target_matches
        )
        ==
        1
    )

    else None
)


# ============================================================
# 10) Resolution failures
# ============================================================

cell73_resolution_failures = []


if (
    cell73_v2_packet
    is None
):

    cell73_resolution_failures.append(
        "V2_PILOT_PACKET_NOT_RESOLVED"
    )


if (
    len(
        cell73_case_matches
    )
    !=
    1
):

    cell73_resolution_failures.append(
        {
            "reason":
                "V2_CASE_RESOLUTION_NOT_UNIQUE",

            "count":
                len(
                    cell73_case_matches
                ),
        }
    )


if (
    len(
        cell73_task_matches
    )
    !=
    1
):

    cell73_resolution_failures.append(
        {
            "reason":
                "V2_TASK_RESOLUTION_NOT_UNIQUE",

            "count":
                len(
                    cell73_task_matches
                ),
        }
    )


if (
    len(
        cell73_target_matches
    )
    !=
    1
):

    cell73_resolution_failures.append(
        {
            "reason":
                "V2_TARGET_RESOLUTION_NOT_UNIQUE",

            "count":
                len(
                    cell73_target_matches
                ),
        }
    )


# ============================================================
# 11) Chain linkage validation
# ============================================================

cell73_chain_failures = []


if (
    cell73_v2_packet
    is not None
):

    for field_name, expected_value in (
        cell73_expected_chain.items()
    ):

        if (
            cell73_v2_packet.get(
                field_name
            )
            !=
            expected_value
        ):

            cell73_chain_failures.append(
                {
                    "object":
                        "pilot_packet",

                    "field":
                        field_name,

                    "expected":
                        expected_value,

                    "actual":
                        cell73_v2_packet.get(
                            field_name
                        ),
                }
            )


if (
    cell73_v2_task
    is not None
):

    if (
        cell73_v2_task.get(
            "golden_case_id"
        )
        !=
        cell73_expected_chain[
            "golden_case_id"
        ]
    ):

        cell73_chain_failures.append(
            "TASK_TO_CASE_LINKAGE_MISMATCH"
        )


if (
    cell73_v2_target
    is not None
):

    if (
        cell73_v2_target.get(
            "golden_case_id"
        )
        !=
        cell73_expected_chain[
            "golden_case_id"
        ]
    ):

        cell73_chain_failures.append(
            "TARGET_TO_CASE_LINKAGE_MISMATCH"
        )


    if (
        cell73_v2_target.get(
            "task_draft_id"
        )
        !=
        cell73_expected_chain[
            "task_draft_id"
        ]
    ):

        cell73_chain_failures.append(
            "TARGET_TO_TASK_LINKAGE_MISMATCH"
        )


# ============================================================
# 12) Source + stratum validation
# ============================================================

cell73_identity_semantic_failures = []


for object_name, record in (
    [
        (
            "packet",
            cell73_v2_packet,
        ),
        (
            "case",
            cell73_v2_case,
        ),
        (
            "task",
            cell73_v2_task,
        ),
        (
            "target",
            cell73_v2_target,
        ),
    ]
):

    if (
        record is None
    ):

        continue


    if (
        record.get(
            "source_type"
        )
        !=
        "book"
    ):

        cell73_identity_semantic_failures.append(
            {
                "object":
                    object_name,

                "reason":
                    "SOURCE_TYPE_NOT_BOOK",

                "actual":
                    record.get(
                        "source_type"
                    ),
            }
        )


    if (
        record.get(
            "stratum"
        )
        !=
        "exercise_or_self_assessment"
    ):

        cell73_identity_semantic_failures.append(
            {
                "object":
                    object_name,

                "reason":
                    "STRATUM_MISMATCH",

                "actual":
                    record.get(
                        "stratum"
                    ),
            }
        )


# ============================================================
# 13) Target parts
#
# Primary inspection text comes from the actual Question
# Drafting packet because Cell70/71 proved that packet builder
# exactly reproduces the validated target representation.
# ============================================================

cell73_target_parts = (
    cell73_v2_packet.get(
        "target_text_parts",
        [],
    )

    if (
        cell73_v2_packet
        is not None
    )

    else []
)


cell73_part_integrity_failures = []


if (
    len(
        cell73_target_parts
    )
    <
    2
):

    cell73_part_integrity_failures.append(
        {
            "reason":
                "TARGET_NOT_MULTIPART",

            "part_count":
                len(
                    cell73_target_parts
                ),
        }
    )


for part_index, part in enumerate(
    cell73_target_parts,
    start=1,
):

    text = (
        part.get(
            "text"
        )
    )


    if not (
        isinstance(
            text,
            str,
        )

        and

        text.strip()
    ):

        cell73_part_integrity_failures.append(
            {
                "part_index":
                    part_index,

                "reason":
                    "EMPTY_TEXT",
            }
        )


    locator = (
        part.get(
            "locator"
        )
    )


    if not (
        isinstance(
            locator,
            dict,
        )

        and

        locator
    ):

        cell73_part_integrity_failures.append(
            {
                "part_index":
                    part_index,

                "reason":
                    "MISSING_LOCATOR",
            }
        )


# ============================================================
# 14) Actual text / token diagnostics
# ============================================================

cell73_target_texts = [
    part.get(
        "text",
        ""
    )

    for part
    in cell73_target_parts
]


cell73_joined_target_text = "\n\n".join(
    cell73_target_texts
)


cell73_joined_target_sha256 = (
    hashlib.sha256(
        cell73_joined_target_text.encode(
            "utf-8"
        )
    ).hexdigest()
)


cell73_part_token_values = [
    part.get(
        "token_count"
    )

    for part
    in cell73_target_parts
]


cell73_numeric_part_token_sum = sum(
    value

    for value
    in cell73_part_token_values

    if (
        isinstance(
            value,
            int,
        )
    )
)


cell73_packet_target_total_tokens = (
    cell73_v2_packet.get(
        "target_total_tokens"
    )
    if (
        cell73_v2_packet
        is not None
    )
    else None
)


cell73_target_record_total_tokens = (
    cell73_v2_target.get(
        "target_total_tokens"
    )
    if (
        cell73_v2_target
        is not None
    )
    else None
)


# Token equality is useful evidence, but not required merely
# to make the packet semantically inspectable.
cell73_token_diagnostics = {
    "packet_target_total_tokens":
        cell73_packet_target_total_tokens,

    "target_record_total_tokens":
        cell73_target_record_total_tokens,

    "sum_of_numeric_packet_part_tokens":
        cell73_numeric_part_token_sum,

    "packet_vs_target_equal":
        (
            cell73_packet_target_total_tokens
            ==
            cell73_target_record_total_tokens
        ),

    "packet_vs_part_sum_equal":
        (
            cell73_packet_target_total_tokens
            ==
            cell73_numeric_part_token_sum
        ),
}


# ============================================================
# 15) Actual semantic contract diagnostics
#
# IMPORTANT:
# These are NOT hard-coded pass/fail assumptions.
#
# We print the real values carried by:
#   Pilot Packet
#   Task
#   Target
#
# so semantic adjudication uses evidence, not guesses.
# ============================================================

CELL73_SEMANTIC_FIELDS = [
    "task_family",
    "question_archetype",
    "required_dimensions",
    "authority_expectation",
    "context_expectation",
    "evidence_scope",
    "target_selection_method",
    "target_selection_basis",
    "evidence_target_method",
    "target_authority_state",
    "target_selection_basis",
]


cell73_semantic_diagnostics = {}


for field_name in (
    CELL73_SEMANTIC_FIELDS
):

    cell73_semantic_diagnostics[
        field_name
    ] = {
        "packet":
            (
                cell73_v2_packet.get(
                    field_name
                )
                if (
                    cell73_v2_packet
                    is not None
                )
                else None
            ),

        "task":
            (
                cell73_v2_task.get(
                    field_name
                )
                if (
                    cell73_v2_task
                    is not None
                )
                else None
            ),

        "target":
            (
                cell73_v2_target.get(
                    field_name
                )
                if (
                    cell73_v2_target
                    is not None
                )
                else None
            ),
    }


# ============================================================
# 16) Heading / locator diagnostics
# ============================================================

cell73_heading_paths = [
    (
        part.get(
            "locator",
            {}
        ).get(
            "heading_path"
        )
    )

    for part
    in cell73_target_parts
]


cell73_document_ids = [
    part.get(
        "document_id"
    )

    for part
    in cell73_target_parts
]


cell73_part_roles = [
    part.get(
        "role"
    )

    for part
    in cell73_target_parts
]


# ============================================================
# 17) Legacy leakage
# ============================================================

cell73_packet_serialized = (
    stable_json_key(
        {
            "packet":
                cell73_v2_packet,

            "case":
                cell73_v2_case,

            "task":
                cell73_v2_task,

            "target":
                cell73_v2_target,
        }
    )
)


cell73_legacy_leakage = {
    "case_ids":
        sorted(
            [
                item_id

                for item_id
                in cell65_legacy_exercise_case_ids

                if (
                    item_id
                    in
                    cell73_packet_serialized
                )
            ]
        ),

    "task_ids":
        sorted(
            [
                item_id

                for item_id
                in cell65_legacy_exercise_task_ids

                if (
                    item_id
                    in
                    cell73_packet_serialized
                )
            ]
        ),

    "target_ids":
        sorted(
            [
                item_id

                for item_id
                in cell65_legacy_exercise_target_ids

                if (
                    item_id
                    in
                    cell73_packet_serialized
                )
            ]
        ),
}


# ============================================================
# 18) Generation state — diagnostic + safety gate
# ============================================================

cell73_generation_state = {}


if (
    cell73_v2_packet
    is not None
):

    for field_name in (
        [
            "question_draft_status",
            "question_groundability_status",
            "expected_answer_status",
            "citation_target_validation_status",
        ]
    ):

        if (
            field_name
            in
            cell73_v2_packet
        ):

            cell73_generation_state[
                field_name
            ] = (
                cell73_v2_packet[
                    field_name
                ]
            )


cell73_generation_state_failures = [
    {
        "field":
            field_name,

        "actual":
            value,
    }

    for field_name, value
    in cell73_generation_state.items()

    if (
        value
        !=
        "NOT_STARTED"
    )
]


# ============================================================
# 19) Machine inspectability
#
# This is deliberately narrower than semantic approval.
#
# Semantic approval comes AFTER reading the actual evidence
# printed by this Cell.
# ============================================================

cell73_machine_inspectability_ready = (
    len(
        cell73_prerequisite_failures
    )
    ==
    0

    and

    len(
        cell73_resolution_failures
    )
    ==
    0

    and

    len(
        cell73_chain_failures
    )
    ==
    0

    and

    len(
        cell73_identity_semantic_failures
    )
    ==
    0

    and

    len(
        cell73_part_integrity_failures
    )
    ==
    0

    and

    len(
        cell73_target_parts
    )
    >=
    2

    and

    bool(
        cell73_joined_target_text.strip()
    )

    and

    not any(
        cell73_legacy_leakage.values()
    )

    and

    len(
        cell73_generation_state_failures
    )
    ==
    0
)


cell73_semantic_adjudication_status = (
    "READY_FOR_FINAL_SEMANTIC_ADJUDICATION"

    if (
        cell73_machine_inspectability_ready
    )

    else
    "SEMANTIC_PACKET_NOT_INSPECTABLE"
)


# ============================================================
# 20) Freeze inspection packet
# ============================================================

EXERCISE_V2_PILOT_SEMANTIC_ADJUDICATION_PACKET_V1_DRAFT = {
    "version":
        "exercise_v2_pilot_semantic_adjudication_packet_v1_draft",

    "status":
        cell73_semantic_adjudication_status,

    "authorization_status":
        CELL73_EXPLICIT_ACTIVATION_AUTHORIZATION,

    "pilot_case_id":
        (
            cell73_v2_packet.get(
                "pilot_case_id"
            )
            if (
                cell73_v2_packet
                is not None
            )
            else None
        ),

    "golden_case_id":
        cell73_expected_chain[
            "golden_case_id"
        ],

    "task_draft_id":
        cell73_expected_chain[
            "task_draft_id"
        ],

    "target_evidence_id":
        cell73_expected_chain[
            "target_evidence_id"
        ],

    "target_part_count":
        len(
            cell73_target_parts
        ),

    "packet_target_total_tokens":
        cell73_packet_target_total_tokens,

    "target_record_total_tokens":
        cell73_target_record_total_tokens,

    "joined_target_sha256":
        cell73_joined_target_sha256,

    "human_semantic_decision":
        "PENDING",

    "activation_execution":
        "NOT_STARTED",

    "question_generation":
        "NOT_STARTED",

    "artifact_write":
        "NONE",
}


# ============================================================
# 21) No-mutation recheck
# ============================================================

cell73_active_snapshot_after = {
    "golden":
        cell73_sha(
            golden_case_records
        ),

    "auto":
        cell73_sha(
            auto_case_records
        ),

    "groundability":
        cell73_sha(
            case_groundability_precheck
        ),

    "task_design":
        cell73_sha(
            case_task_design
        ),

    "tasks":
        cell73_sha(
            task_draft_plan_rows
        ),

    "targets":
        cell73_sha(
            target_evidence_records
        ),

    "pilots":
        cell73_sha(
            pilot_selection_rows
        ),
}


# ============================================================
# 22) Gates
# ============================================================

cell73_gates = {
    "explicit_activation_authorization_is_recorded":
        (
            CELL73_EXPLICIT_ACTIVATION_AUTHORIZATION
            ==
            "PROVIDED_BY_USER_IN_CHAT"
        ),

    "cell72_and_cell71_preconditions_remain_valid":
        (
            len(
                cell73_prerequisite_failures
            )
            ==
            0
        ),

    "v2_pilot_case_task_target_resolve_uniquely":
        (
            len(
                cell73_resolution_failures
            )
            ==
            0
        ),

    "v2_pilot_case_task_target_chain_is_exact":
        (
            len(
                cell73_chain_failures
            )
            ==
            0
        ),

    "all_resolved_objects_are_book_exercise_or_self_assessment":
        (
            len(
                cell73_identity_semantic_failures
            )
            ==
            0
        ),

    "target_evidence_is_multirecord_nonempty_and_fully_located":
        (
            len(
                cell73_part_integrity_failures
            )
            ==
            0

            and

            len(
                cell73_target_parts
            )
            >=
            2
        ),

    "joined_target_text_is_nonempty":
        (
            bool(
                cell73_joined_target_text.strip()
            )
        ),

    "semantic_inspection_bundle_has_zero_legacy_leakage":
        (
            not any(
                cell73_legacy_leakage.values()
            )
        ),

    "question_and_answer_generation_are_not_started":
        (
            len(
                cell73_generation_state_failures
            )
            ==
            0
        ),

    "packet_is_ready_for_real_semantic_adjudication":
        (
            cell73_semantic_adjudication_status
            ==
            "READY_FOR_FINAL_SEMANTIC_ADJUDICATION"
        ),

    "live_active_state_remains_byte_stable":
        (
            cell73_active_snapshot_before
            ==
            cell73_active_snapshot_after
        ),

    "actual_activation_not_started":
        (
            EXERCISE_V2_PILOT_SEMANTIC_ADJUDICATION_PACKET_V1_DRAFT[
                "activation_execution"
            ]
            ==
            "NOT_STARTED"
        ),

    "no_artifact_write":
        (
            EXERCISE_V2_PILOT_SEMANTIC_ADJUDICATION_PACKET_V1_DRAFT[
                "artifact_write"
            ]
            ==
            "NONE"
        ),
}


cell_73_pass = all(
    cell73_gates.values()
)


# ============================================================
# 23) Output
# ============================================================

print(
    "=== EXERCISE V2 PILOT — SEMANTIC INSPECTION REPAIRED ==="
)


print(
    "\n=== AUTHORIZATION ==="
)


print(
    CELL73_EXPLICIT_ACTIVATION_AUTHORIZATION
)


print(
    "\n=== RESOLUTION ==="
)


print(
    "Prerequisite failures:",
    cell73_prerequisite_failures
)


print(
    "Resolution failures:",
    cell73_resolution_failures
)


print(
    "Chain failures:",
    cell73_chain_failures
)


print(
    "Identity/stratum failures:",
    cell73_identity_semantic_failures
)


print(
    "\n=== EXACT V2 CHAIN ==="
)


print(
    "question-pilot:",
    (
        cell73_v2_packet.get(
            "pilot_case_id"
        )
        if (
            cell73_v2_packet
            is not None
        )
        else None
    )
)


print(
    "golden_case_id:",
    cell73_expected_chain[
        "golden_case_id"
    ]
)


print(
    "task_draft_id:",
    cell73_expected_chain[
        "task_draft_id"
    ]
)


print(
    "target_evidence_id:",
    cell73_expected_chain[
        "target_evidence_id"
    ]
)


print(
    "\n=== ACTUAL SEMANTIC CONTRACT VALUES ==="
)


for field_name, values in (
    cell73_semantic_diagnostics.items()
):

    print(
        "\nFIELD:",
        field_name
    )


    print(
        "  Packet:",
        values[
            "packet"
        ]
    )


    print(
        "  Task:",
        values[
            "task"
        ]
    )


    print(
        "  Target:",
        values[
            "target"
        ]
    )


print(
    "\n=== TARGET STRUCTURE ==="
)


print(
    "Part count:",
    len(
        cell73_target_parts
    )
)


print(
    "Part roles:",
    cell73_part_roles
)


print(
    "Document IDs:",
    cell73_document_ids
)


print(
    "Heading paths:",
    cell73_heading_paths
)


print(
    "Token diagnostics:",
    cell73_token_diagnostics
)


print(
    "Part integrity failures:",
    cell73_part_integrity_failures
)


print(
    "\n=== FULL TARGET EVIDENCE ==="
)


for part_index, part in enumerate(
    cell73_target_parts,
    start=1,
):

    print(
        "\n----------------------------------------"
    )


    print(
        "PART",
        part_index
    )


    print(
        "Role:",
        part.get(
            "role"
        )
    )


    print(
        "Document ID:",
        part.get(
            "document_id"
        )
    )


    print(
        "Token count:",
        part.get(
            "token_count"
        )
    )


    print(
        "Locator:",
        part.get(
            "locator"
        )
    )


    print(
        "Attribution:",
        part.get(
            "attribution"
        )
    )


    print(
        "\nTEXT:"
    )


    print(
        part.get(
            "text"
        )
    )


print(
    "\n========================================"
)


print(
    "\n=== JOINED TARGET TEXT ==="
)


print(
    cell73_joined_target_text
)


print(
    "\n=== SAFETY / STATE ==="
)


print(
    "Legacy leakage:",
    cell73_legacy_leakage
)


print(
    "Generation state:",
    cell73_generation_state
)


print(
    "Generation failures:",
    cell73_generation_state_failures
)


print(
    "\nJoined target SHA256:",
    cell73_joined_target_sha256
)


print(
    "\nSEMANTIC ADJUDICATION STATUS:",
    cell73_semantic_adjudication_status
)


print(
    "\n=== CODE CELL 73 GATES ==="
)


for gate_name, passed in (
    cell73_gates.items()
):

    print(
        gate_name,
        ":",
        "PASS"
        if passed
        else "FAIL"
    )


print(
    "\nFAILED GATES:",
    [
        gate_name

        for gate_name, passed
        in cell73_gates.items()

        if not passed
    ]
)


print(
    "\nCODE CELL 73 GATE:",
    "PASS"
    if cell_73_pass
    else "FAIL"
)


assert cell_73_pass, (
    "The V2 Exercise pilot still cannot be inspected safely. "
    "Do not activate. Use the explicit resolution, linkage, "
    "target-part and generation-state diagnostics from this "
    "SAME Code Cell 73."
)
```

```python
# Code Cell 74

# ============================================================
# AUTHORIZED ATOMIC ACTIVATION — EXERCISE V2
#
# This is the FIRST Cell that intentionally mutates the
# canonical Active In-Memory State.
#
# Authorization:
#   explicitly provided by user in chat.
#
# Semantic decision:
#   PASS
#
# Transaction:
#   7 canonical registries
#   4 canonical indexes
#   5 true current-state derived objects
#
# Failure policy:
#   AUTOMATIC ROLLBACK TO EXACT ORIGINAL OBJECT REFERENCES
#
# NO file write.
# NO Commit.
# NO Push.
# NO PR.
# NO question generation.
# ============================================================


# ============================================================
# 1) Helpers
# ============================================================

def cell74_sha(
    value,
):

    return hashlib.sha256(
        stable_json_key(
            value
        ).encode(
            "utf-8"
        )
    ).hexdigest()


def cell74_clone(
    value,
):

    return cell52_safe_clone(
        value
    )


# ============================================================
# 2) Explicit authorization + semantic decision
# ============================================================

CELL74_ACTIVATION_AUTHORIZATION = (
    "PROVIDED_BY_USER_IN_CHAT"
)


CELL74_SEMANTIC_DECISION = (
    "PASS"
)


EXERCISE_V2_PILOT_SEMANTIC_DECISION_V1 = {
    "version":
        "exercise_v2_pilot_semantic_decision_v1",

    "decision":
        CELL74_SEMANTIC_DECISION,

    "decision_source":
        "ASSISTANT_ADJUDICATION_FROM_CELL73_PRINTED_EVIDENCE",

    "evidence_joined_target_sha256":
        cell73_joined_target_sha256,

    "evidence_part_count":
        len(
            cell73_target_parts
        ),

    "reason_codes": [
        "EXPLICIT_SELF_ASSESSMENT_TITLE",
        "DIRECT_LEARNER_SELF_EVALUATION_PROMPT",
        "WRITTEN_RESPONSE_INSTRUCTION",
        "FIVE_LEVEL_STAIRCASE_SCALE_INSTRUCTION",
        "COMPLETE_MULTI_RECORD_EXERCISE_BOUNDARY",
    ],

    "authorization":
        CELL74_ACTIVATION_AUTHORIZATION,
}


# ============================================================
# 3) Preconditions
# ============================================================

cell74_prerequisite_failures = []


if not (
    cell_73_pass
):

    cell74_prerequisite_failures.append(
        "CELL73_NOT_PASS"
    )


if (
    cell73_semantic_adjudication_status
    !=
    "READY_FOR_FINAL_SEMANTIC_ADJUDICATION"
):

    cell74_prerequisite_failures.append(
        "CELL73_SEMANTIC_PACKET_NOT_READY"
    )


if (
    CELL74_SEMANTIC_DECISION
    !=
    "PASS"
):

    cell74_prerequisite_failures.append(
        "SEMANTIC_DECISION_NOT_PASS"
    )


if (
    CELL74_ACTIVATION_AUTHORIZATION
    !=
    "PROVIDED_BY_USER_IN_CHAT"
):

    cell74_prerequisite_failures.append(
        "ACTIVATION_AUTHORIZATION_NOT_PROVIDED"
    )


if not (
    cell72_technical_activation_ready
):

    cell74_prerequisite_failures.append(
        "CELL72_TECHNICAL_READINESS_FALSE"
    )


if not (
    cell71_full_transaction_clone_ready
):

    cell74_prerequisite_failures.append(
        "CELL71_FULL_TRANSACTION_CLONE_NOT_READY"
    )


# ============================================================
# 4) Exact BEFORE-state verification
#
# Activation is allowed ONLY if current live state still
# equals the exact BEFORE fingerprint validated in Cell67.
# ============================================================

cell74_live_before_registry_payload = {
    "golden_case_records":
        golden_case_records,

    "auto_case_records":
        auto_case_records,

    "case_groundability_precheck":
        case_groundability_precheck,

    "case_task_design":
        case_task_design,

    "task_draft_plan_rows":
        task_draft_plan_rows,

    "target_evidence_records":
        target_evidence_records,

    "pilot_selection_rows":
        pilot_selection_rows,
}


cell74_live_before_index_payload = {
    "golden_case_records_by_id":
        golden_case_records_by_id,

    "task_draft_plans":
        task_draft_plans,

    "target_evidence_by_case_id":
        target_evidence_by_case_id,

    "target_evidence_by_task_draft_id":
        target_evidence_by_task_draft_id,
}


cell74_live_before_bundle_sha256 = (
    cell74_sha(
        {
            "registries":
                cell74_live_before_registry_payload,

            "indexes":
                cell74_live_before_index_payload,
        }
    )
)


if (
    cell74_live_before_bundle_sha256
    !=
    cell67_before_bundle_fingerprint
):

    cell74_prerequisite_failures.append(
        {
            "reason":
                "LIVE_STATE_DRIFTED_FROM_VALIDATED_BEFORE",

            "expected":
                cell67_before_bundle_fingerprint,

            "actual":
                cell74_live_before_bundle_sha256,
        }
    )


# ============================================================
# 5) Preserve EXACT ORIGINAL OBJECT REFERENCES for rollback
#
# We intentionally keep references rather than deep clones.
# The transaction rebinds variable names; it does not mutate
# these old objects in place.
#
# This preserves alias relationships during rollback.
# ============================================================

cell74_before_object_refs = {
    # Canonical registries
    "golden_case_records":
        golden_case_records,

    "auto_case_records":
        auto_case_records,

    "case_groundability_precheck":
        case_groundability_precheck,

    "case_task_design":
        case_task_design,

    "task_draft_plan_rows":
        task_draft_plan_rows,

    "target_evidence_records":
        target_evidence_records,

    "pilot_selection_rows":
        pilot_selection_rows,

    # Canonical indexes
    "golden_case_records_by_id":
        golden_case_records_by_id,

    "task_draft_plans":
        task_draft_plans,

    "target_evidence_by_case_id":
        target_evidence_by_case_id,

    "target_evidence_by_task_draft_id":
        target_evidence_by_task_draft_id,

    # Five true derived objects
    "candidate_hash_to_case_ids":
        candidate_hash_to_case_ids,

    "cross_stratum_repeated_candidate_hashes":
        cross_stratum_repeated_candidate_hashes,

    "question_drafting_pilot_packets":
        question_drafting_pilot_packets,

    "question_drafting_pilot_by_id":
        question_drafting_pilot_by_id,

    "target_id_counts":
        target_id_counts,
}


# ============================================================
# 6) Snapshot Historical + SAFE objects
#
# They MUST NOT change during activation.
# ============================================================

cell74_historical_snapshot_before = {
    variable_name:
        cell74_sha(
            globals()[
                variable_name
            ]
        )

    for variable_name
    in cell69_v2_historical_preserve_objects
}


cell74_safe_snapshot_before = {
    variable_name:
        cell74_sha(
            globals()[
                variable_name
            ]
        )

    for variable_name
    in cell69_v2_safe_no_action_objects
}


# ============================================================
# 7) Rollback helper
# ============================================================

def cell74_restore_exact_before_state():

    global golden_case_records
    global auto_case_records
    global case_groundability_precheck
    global case_task_design
    global task_draft_plan_rows
    global target_evidence_records
    global pilot_selection_rows

    global golden_case_records_by_id
    global task_draft_plans
    global target_evidence_by_case_id
    global target_evidence_by_task_draft_id

    global candidate_hash_to_case_ids
    global cross_stratum_repeated_candidate_hashes
    global question_drafting_pilot_packets
    global question_drafting_pilot_by_id
    global target_id_counts


    golden_case_records = (
        cell74_before_object_refs[
            "golden_case_records"
        ]
    )


    auto_case_records = (
        cell74_before_object_refs[
            "auto_case_records"
        ]
    )


    case_groundability_precheck = (
        cell74_before_object_refs[
            "case_groundability_precheck"
        ]
    )


    case_task_design = (
        cell74_before_object_refs[
            "case_task_design"
        ]
    )


    task_draft_plan_rows = (
        cell74_before_object_refs[
            "task_draft_plan_rows"
        ]
    )


    target_evidence_records = (
        cell74_before_object_refs[
            "target_evidence_records"
        ]
    )


    pilot_selection_rows = (
        cell74_before_object_refs[
            "pilot_selection_rows"
        ]
    )


    golden_case_records_by_id = (
        cell74_before_object_refs[
            "golden_case_records_by_id"
        ]
    )


    task_draft_plans = (
        cell74_before_object_refs[
            "task_draft_plans"
        ]
    )


    target_evidence_by_case_id = (
        cell74_before_object_refs[
            "target_evidence_by_case_id"
        ]
    )


    target_evidence_by_task_draft_id = (
        cell74_before_object_refs[
            "target_evidence_by_task_draft_id"
        ]
    )


    candidate_hash_to_case_ids = (
        cell74_before_object_refs[
            "candidate_hash_to_case_ids"
        ]
    )


    cross_stratum_repeated_candidate_hashes = (
        cell74_before_object_refs[
            "cross_stratum_repeated_candidate_hashes"
        ]
    )


    question_drafting_pilot_packets = (
        cell74_before_object_refs[
            "question_drafting_pilot_packets"
        ]
    )


    question_drafting_pilot_by_id = (
        cell74_before_object_refs[
            "question_drafting_pilot_by_id"
        ]
    )


    target_id_counts = (
        cell74_before_object_refs[
            "target_id_counts"
        ]
    )


# ============================================================
# 8) Transaction execution state
# ============================================================

cell74_transaction_exception = None

cell74_post_activation_failures = []

cell74_rollback_executed = False

cell74_activation_execution_started = False

cell74_activation_committed = False


# ============================================================
# 9) Execute ONLY if all preconditions pass
# ============================================================

if (
    len(
        cell74_prerequisite_failures
    )
    ==
    0
):

    cell74_activation_execution_started = True


    try:

        # ====================================================
        # PHASE A — Canonical Registry Swap
        # ====================================================

        cell74_after_registry_clone = (
            cell74_clone(
                cell67_after_state
            )
        )


        golden_case_records = (
            cell74_after_registry_clone[
                "golden_case_records"
            ]
        )


        auto_case_records = (
            cell74_after_registry_clone[
                "auto_case_records"
            ]
        )


        case_groundability_precheck = (
            cell74_after_registry_clone[
                "case_groundability_precheck"
            ]
        )


        case_task_design = (
            cell74_after_registry_clone[
                "case_task_design"
            ]
        )


        task_draft_plan_rows = (
            cell74_after_registry_clone[
                "task_draft_plan_rows"
            ]
        )


        target_evidence_records = (
            cell74_after_registry_clone[
                "target_evidence_records"
            ]
        )


        pilot_selection_rows = (
            cell74_after_registry_clone[
                "pilot_selection_rows"
            ]
        )


        # ====================================================
        # PHASE B — Rebuild four Canonical Indexes from the
        # NEW LIVE registries.
        #
        # Do NOT independently deep-clone prebuilt indexes.
        # Rebuilding preserves normal row-reference behavior.
        # ====================================================

        golden_case_records_by_id = {
            row[
                "golden_case_id"
            ]:
                row

            for row
            in golden_case_records
        }


        task_draft_plans = {
            row[
                "task_draft_id"
            ]:
                row

            for row
            in task_draft_plan_rows
        }


        target_evidence_by_case_id = {
            row[
                "golden_case_id"
            ]:
                row

            for row
            in target_evidence_records
        }


        target_evidence_by_task_draft_id = {
            row[
                "task_draft_id"
            ]:
                row

            for row
            in target_evidence_records
        }


        # ====================================================
        # PHASE C — Rebuild three simple Current-State
        # derivatives from NEW LIVE state.
        # ====================================================

        cell74_new_candidate_hash_to_case_ids = defaultdict(
            list
        )


        for case_record in (
            golden_case_records
        ):

            cell74_new_candidate_hash_to_case_ids[
                case_record[
                    "candidate_identity_sha256"
                ]
            ].append(
                case_record[
                    "golden_case_id"
                ]
            )


        cell74_new_cross_stratum_repeated_candidate_hashes = {
            candidate_hash:
                case_ids

            for candidate_hash, case_ids
            in cell74_new_candidate_hash_to_case_ids.items()

            if (
                len(
                    case_ids
                )
                >
                1
            )
        }


        cell74_new_target_id_counts = Counter(
            record[
                "target_evidence_id"
            ]

            for record
            in target_evidence_records
        )


        # ====================================================
        # PHASE D — Build Question Drafting packet/index using
        # EXACT Cell57 source block already proven by Cell70
        # and Cell71.
        # ====================================================

        cell74_new_targets_by_stratum = defaultdict(
            list
        )


        for target_record in (
            target_evidence_records
        ):

            cell74_new_targets_by_stratum[
                (
                    target_record[
                        "source_type"
                    ],
                    target_record[
                        "stratum"
                    ],
                )
            ].append(
                target_record
            )


        cell74_packet_exec_namespace = dict(
            globals()
        )


        cell74_packet_exec_namespace.update(
            {
                "golden_case_records":
                    golden_case_records,

                "golden_case_records_by_id":
                    golden_case_records_by_id,

                "auto_case_records":
                    auto_case_records,

                "case_groundability_precheck":
                    case_groundability_precheck,

                "case_task_design":
                    case_task_design,

                "task_draft_plan_rows":
                    task_draft_plan_rows,

                "task_draft_plans":
                    task_draft_plans,

                "target_evidence_records":
                    target_evidence_records,

                "target_evidence_by_case_id":
                    target_evidence_by_case_id,

                "target_evidence_by_task_draft_id":
                    target_evidence_by_task_draft_id,

                "pilot_selection_rows":
                    pilot_selection_rows,

                "targets_by_stratum":
                    cell74_new_targets_by_stratum,
            }
        )


        exec(
            cell70_packet_build_source,
            cell74_packet_exec_namespace,
            cell74_packet_exec_namespace,
        )


        cell74_new_question_drafting_pilot_packets = (
            cell74_packet_exec_namespace[
                "question_drafting_pilot_packets"
            ]
        )


        cell74_new_question_drafting_pilot_by_id = (
            cell74_packet_exec_namespace[
                "question_drafting_pilot_by_id"
            ]
        )


        # ====================================================
        # PHASE E — Bind five rebuilt Current-State derived
        # objects to ACTIVE names.
        # ====================================================

        candidate_hash_to_case_ids = (
            cell74_new_candidate_hash_to_case_ids
        )


        cross_stratum_repeated_candidate_hashes = (
            cell74_new_cross_stratum_repeated_candidate_hashes
        )


        target_id_counts = (
            cell74_new_target_id_counts
        )


        question_drafting_pilot_packets = (
            cell74_new_question_drafting_pilot_packets
        )


        question_drafting_pilot_by_id = (
            cell74_new_question_drafting_pilot_by_id
        )


        # ====================================================
        # PHASE F — Post-Activation Validation
        # ====================================================

        cell74_live_after_registry_payload = {
            "golden_case_records":
                golden_case_records,

            "auto_case_records":
                auto_case_records,

            "case_groundability_precheck":
                case_groundability_precheck,

            "case_task_design":
                case_task_design,

            "task_draft_plan_rows":
                task_draft_plan_rows,

            "target_evidence_records":
                target_evidence_records,

            "pilot_selection_rows":
                pilot_selection_rows,
        }


        cell74_live_after_index_payload = {
            "golden_case_records_by_id":
                golden_case_records_by_id,

            "task_draft_plans":
                task_draft_plans,

            "target_evidence_by_case_id":
                target_evidence_by_case_id,

            "target_evidence_by_task_draft_id":
                target_evidence_by_task_draft_id,
        }


        cell74_live_after_derived_payload = {
            "candidate_hash_to_case_ids":
                candidate_hash_to_case_ids,

            "cross_stratum_repeated_candidate_hashes":
                cross_stratum_repeated_candidate_hashes,

            "question_drafting_pilot_packets":
                question_drafting_pilot_packets,

            "question_drafting_pilot_by_id":
                question_drafting_pilot_by_id,

            "target_id_counts":
                target_id_counts,
        }


        # ----------------------------------------------------
        # F1 — Exact canonical AFTER fingerprint
        # ----------------------------------------------------

        cell74_live_after_canonical_sha256 = (
            cell74_sha(
                {
                    "registries":
                        cell74_live_after_registry_payload,

                    "indexes":
                        cell74_live_after_index_payload,
                }
            )
        )


        if (
            cell74_live_after_canonical_sha256
            !=
            cell67_after_bundle_fingerprint
        ):

            cell74_post_activation_failures.append(
                {
                    "reason":
                        "LIVE_AFTER_CANONICAL_FINGERPRINT_MISMATCH",

                    "expected":
                        cell67_after_bundle_fingerprint,

                    "actual":
                        cell74_live_after_canonical_sha256,
                }
            )


        # ----------------------------------------------------
        # F2 — Exact derived AFTER fingerprint
        # ----------------------------------------------------

        cell74_live_after_derived_sha256 = (
            cell74_sha(
                cell74_live_after_derived_payload
            )
        )


        if (
            cell74_live_after_derived_sha256
            !=
            cell71_after_derived_bundle_sha256
        ):

            cell74_post_activation_failures.append(
                {
                    "reason":
                        "LIVE_AFTER_DERIVED_FINGERPRINT_MISMATCH",

                    "expected":
                        cell71_after_derived_bundle_sha256,

                    "actual":
                        cell74_live_after_derived_sha256,
                }
            )


        # ----------------------------------------------------
        # F3 — Exact canonical counts
        # ----------------------------------------------------

        cell74_live_counts = {
            "golden":
                len(
                    golden_case_records
                ),

            "auto":
                len(
                    auto_case_records
                ),

            "groundability":
                len(
                    case_groundability_precheck
                ),

            "task_design":
                len(
                    case_task_design
                ),

            "tasks":
                len(
                    task_draft_plan_rows
                ),

            "targets":
                len(
                    target_evidence_records
                ),

            "pilots":
                len(
                    pilot_selection_rows
                ),

            "question_packets":
                len(
                    question_drafting_pilot_packets
                ),
        }


        CELL74_EXPECTED_COUNTS = {
            "golden":
                724,

            "auto":
                572,

            "groundability":
                724,

            "task_design":
                724,

            "tasks":
                724,

            "targets":
                572,

            "pilots":
                23,

            "question_packets":
                23,
        }


        if (
            cell74_live_counts
            !=
            CELL74_EXPECTED_COUNTS
        ):

            cell74_post_activation_failures.append(
                {
                    "reason":
                        "LIVE_AFTER_COUNT_MISMATCH",

                    "expected":
                        CELL74_EXPECTED_COUNTS,

                    "actual":
                        cell74_live_counts,
                }
            )


        # ----------------------------------------------------
        # F4 — Legacy leakage
        # ----------------------------------------------------

        cell74_live_current_state_serialized = (
            stable_json_key(
                {
                    "registries":
                        cell74_live_after_registry_payload,

                    "indexes":
                        cell74_live_after_index_payload,

                    "derived":
                        cell74_live_after_derived_payload,
                }
            )
        )


        cell74_live_legacy_leakage = {
            "case_ids":
                sorted(
                    [
                        item_id

                        for item_id
                        in cell65_legacy_exercise_case_ids

                        if (
                            item_id
                            in
                            cell74_live_current_state_serialized
                        )
                    ]
                ),

            "task_ids":
                sorted(
                    [
                        item_id

                        for item_id
                        in cell65_legacy_exercise_task_ids

                        if (
                            item_id
                            in
                            cell74_live_current_state_serialized
                        )
                    ]
                ),

            "target_ids":
                sorted(
                    [
                        item_id

                        for item_id
                        in cell65_legacy_exercise_target_ids

                        if (
                            item_id
                            in
                            cell74_live_current_state_serialized
                        )
                    ]
                ),
        }


        if (
            any(
                cell74_live_legacy_leakage.values()
            )
        ):

            cell74_post_activation_failures.append(
                {
                    "reason":
                        "LIVE_LEGACY_ID_LEAKAGE",

                    "detail":
                        cell74_live_legacy_leakage,
                }
            )


        # ----------------------------------------------------
        # F5 — V2 admission
        # ----------------------------------------------------

        cell74_live_case_ids = {
            row[
                "golden_case_id"
            ]

            for row
            in golden_case_records
        }


        cell74_live_task_ids = {
            row[
                "task_draft_id"
            ]

            for row
            in task_draft_plan_rows
        }


        cell74_live_target_ids = {
            row[
                "target_evidence_id"
            ]

            for row
            in target_evidence_records
        }


        cell74_live_v2_admission_failures = {
            "case_ids":
                sorted(
                    cell65_v2_case_ids
                    -
                    cell74_live_case_ids
                ),

            "task_ids":
                sorted(
                    cell65_v2_task_ids
                    -
                    cell74_live_task_ids
                ),

            "target_ids":
                sorted(
                    cell65_v2_target_ids
                    -
                    cell74_live_target_ids
                ),
        }


        if (
            any(
                cell74_live_v2_admission_failures.values()
            )
        ):

            cell74_post_activation_failures.append(
                {
                    "reason":
                        "LIVE_V2_ADMISSION_FAILURE",

                    "detail":
                        cell74_live_v2_admission_failures,
                }
            )


        # ----------------------------------------------------
        # F6 — Exercise Pilot Selection
        # ----------------------------------------------------

        cell74_live_exercise_selections = [
            row

            for row
            in pilot_selection_rows

            if (
                row[
                    "source_type"
                ]
                ==
                "book"

                and

                row[
                    "stratum"
                ]
                ==
                "exercise_or_self_assessment"
            )
        ]


        if (
            len(
                cell74_live_exercise_selections
            )
            !=
            1
        ):

            cell74_post_activation_failures.append(
                {
                    "reason":
                        "LIVE_EXERCISE_SELECTION_COUNT_NOT_1",

                    "actual":
                        len(
                            cell74_live_exercise_selections
                        ),
                }
            )


        else:

            cell74_live_exercise_selection = (
                cell74_live_exercise_selections[
                    0
                ]
            )


            for field_name in (
                [
                    "golden_case_id",
                    "task_draft_id",
                    "target_evidence_id",
                ]
            ):

                if (
                    cell74_live_exercise_selection[
                        field_name
                    ]
                    !=
                    staged_exercise_v2_pilot_replacement_row[
                        field_name
                    ]
                ):

                    cell74_post_activation_failures.append(
                        {
                            "reason":
                                "LIVE_EXERCISE_SELECTION_LINKAGE_MISMATCH",

                            "field":
                                field_name,

                            "expected":
                                staged_exercise_v2_pilot_replacement_row[
                                    field_name
                                ],

                            "actual":
                                cell74_live_exercise_selection[
                                    field_name
                                ],
                        }
                    )


        # ----------------------------------------------------
        # F7 — Exercise Question Packet
        # ----------------------------------------------------

        cell74_live_exercise_packets = [
            packet

            for packet
            in question_drafting_pilot_packets

            if (
                packet[
                    "source_type"
                ]
                ==
                "book"

                and

                packet[
                    "stratum"
                ]
                ==
                "exercise_or_self_assessment"
            )
        ]


        if (
            len(
                cell74_live_exercise_packets
            )
            !=
            1
        ):

            cell74_post_activation_failures.append(
                {
                    "reason":
                        "LIVE_EXERCISE_PACKET_COUNT_NOT_1",

                    "actual":
                        len(
                            cell74_live_exercise_packets
                        ),
                }
            )


        else:

            cell74_live_exercise_packet = (
                cell74_live_exercise_packets[
                    0
                ]
            )


            for field_name in (
                [
                    "golden_case_id",
                    "task_draft_id",
                    "target_evidence_id",
                ]
            ):

                if (
                    cell74_live_exercise_packet[
                        field_name
                    ]
                    !=
                    staged_exercise_v2_pilot_replacement_row[
                        field_name
                    ]
                ):

                    cell74_post_activation_failures.append(
                        {
                            "reason":
                                "LIVE_EXERCISE_PACKET_LINKAGE_MISMATCH",

                            "field":
                                field_name,

                            "expected":
                                staged_exercise_v2_pilot_replacement_row[
                                    field_name
                                ],

                            "actual":
                                cell74_live_exercise_packet[
                                    field_name
                                ],
                        }
                    )


            if (
                cell74_live_exercise_packet[
                    "pilot_case_id"
                ]
                !=
                cell71_v2_exercise_packet_id
            ):

                cell74_post_activation_failures.append(
                    {
                        "reason":
                            "LIVE_V2_QUESTION_PILOT_ID_MISMATCH",

                        "expected":
                            cell71_v2_exercise_packet_id,

                        "actual":
                            cell74_live_exercise_packet[
                                "pilot_case_id"
                            ],
                    }
                )


        # ----------------------------------------------------
        # F8 — Old question-pilot identity gone
        # ----------------------------------------------------

        if (
            cell71_legacy_exercise_packet_id
            in
            question_drafting_pilot_by_id
        ):

            cell74_post_activation_failures.append(
                "LEGACY_QUESTION_PILOT_ID_STILL_ACTIVE"
            )


        if (
            cell71_v2_exercise_packet_id
            not in
            question_drafting_pilot_by_id
        ):

            cell74_post_activation_failures.append(
                "V2_QUESTION_PILOT_ID_NOT_ACTIVE"
            )


        # ----------------------------------------------------
        # F9 — Historical + SAFE state untouched
        # ----------------------------------------------------

        cell74_historical_snapshot_after = {
            variable_name:
                cell74_sha(
                    globals()[
                        variable_name
                    ]
                )

            for variable_name
            in cell69_v2_historical_preserve_objects
        }


        cell74_safe_snapshot_after = {
            variable_name:
                cell74_sha(
                    globals()[
                        variable_name
                    ]
                )

            for variable_name
            in cell69_v2_safe_no_action_objects
        }


        if (
            cell74_historical_snapshot_before
            !=
            cell74_historical_snapshot_after
        ):

            cell74_post_activation_failures.append(
                "HISTORICAL_OBJECT_MUTATION_DETECTED"
            )


        if (
            cell74_safe_snapshot_before
            !=
            cell74_safe_snapshot_after
        ):

            cell74_post_activation_failures.append(
                "SAFE_NO_ACTION_OBJECT_MUTATION_DETECTED"
            )


        # ----------------------------------------------------
        # F10 — Full transaction fingerprint
        # ----------------------------------------------------

        cell74_live_full_transaction_payload = {
            "canonical_registries":
                cell74_live_after_registry_payload,

            "canonical_indexes":
                cell74_live_after_index_payload,

            "rebuilt_current_state_derivatives":
                cell74_live_after_derived_payload,

            "historical_preserve_policy":
                {
                    variable_name:
                        "PRESERVE_EXISTING_RUNTIME_OBJECT"

                    for variable_name
                    in sorted(
                        cell69_v2_historical_preserve_objects
                    )
                },

            "safe_no_action_policy":
                {
                    variable_name:
                        "NO_TRANSACTION_ACTION"

                    for variable_name
                    in sorted(
                        cell69_v2_safe_no_action_objects
                    )
                },
        }


        cell74_live_full_transaction_sha256 = (
            cell74_sha(
                cell74_live_full_transaction_payload
            )
        )


        if (
            cell74_live_full_transaction_sha256
            !=
            cell71_full_after_transaction_sha256
        ):

            cell74_post_activation_failures.append(
                {
                    "reason":
                        "FULL_LIVE_TRANSACTION_FINGERPRINT_MISMATCH",

                    "expected":
                        cell71_full_after_transaction_sha256,

                    "actual":
                        cell74_live_full_transaction_sha256,
                }
            )


        # ====================================================
        # PHASE G — Commit-or-Rollback decision
        # ====================================================

        if (
            len(
                cell74_post_activation_failures
            )
            ==
            0
        ):

            cell74_activation_committed = True


        else:

            cell74_restore_exact_before_state()

            cell74_rollback_executed = True


    except Exception as exc:

        cell74_transaction_exception = {
            "error_type":
                type(
                    exc
                ).__name__,

            "error_message":
                str(
                    exc
                ),
        }


        cell74_restore_exact_before_state()

        cell74_rollback_executed = True


# ============================================================
# 10) If preconditions fail, activation never begins
# ============================================================

else:

    cell74_activation_execution_started = False

    cell74_activation_committed = False


# ============================================================
# 11) Final state after commit OR rollback
# ============================================================

cell74_final_live_registry_payload = {
    "golden_case_records":
        golden_case_records,

    "auto_case_records":
        auto_case_records,

    "case_groundability_precheck":
        case_groundability_precheck,

    "case_task_design":
        case_task_design,

    "task_draft_plan_rows":
        task_draft_plan_rows,

    "target_evidence_records":
        target_evidence_records,

    "pilot_selection_rows":
        pilot_selection_rows,
}


cell74_final_live_index_payload = {
    "golden_case_records_by_id":
        golden_case_records_by_id,

    "task_draft_plans":
        task_draft_plans,

    "target_evidence_by_case_id":
        target_evidence_by_case_id,

    "target_evidence_by_task_draft_id":
        target_evidence_by_task_draft_id,
}


cell74_final_live_canonical_sha256 = (
    cell74_sha(
        {
            "registries":
                cell74_final_live_registry_payload,

            "indexes":
                cell74_final_live_index_payload,
        }
    )
)


cell74_final_live_derived_payload = {
    "candidate_hash_to_case_ids":
        candidate_hash_to_case_ids,

    "cross_stratum_repeated_candidate_hashes":
        cross_stratum_repeated_candidate_hashes,

    "question_drafting_pilot_packets":
        question_drafting_pilot_packets,

    "question_drafting_pilot_by_id":
        question_drafting_pilot_by_id,

    "target_id_counts":
        target_id_counts,
}


cell74_final_live_derived_sha256 = (
    cell74_sha(
        cell74_final_live_derived_payload
    )
)


# ============================================================
# 12) Rollback verification
# ============================================================

cell74_rollback_verification_pass = (
    True
)


if (
    cell74_rollback_executed
):

    cell74_rollback_verification_pass = (
        cell74_final_live_canonical_sha256
        ==
        cell67_before_bundle_fingerprint
    )


# ============================================================
# 13) Final Activation Contract
# ============================================================

cell74_activation_status = (
    "ACTIVATED_IN_MEMORY_AND_VALIDATED"

    if (
        cell74_activation_committed

        and

        not cell74_rollback_executed

        and

        cell74_transaction_exception
        is None

        and

        len(
            cell74_post_activation_failures
        )
        ==
        0
    )

    else
    (
        "ROLLED_BACK_TO_BEFORE"

        if (
            cell74_rollback_executed
        )

        else
        "ACTIVATION_NOT_EXECUTED"
    )
)


EXERCISE_V2_ACTIVATION_EXECUTION_V1 = {
    "version":
        "exercise_v2_activation_execution_v1",

    "authorization":
        CELL74_ACTIVATION_AUTHORIZATION,

    "semantic_decision":
        CELL74_SEMANTIC_DECISION,

    "semantic_evidence_sha256":
        cell73_joined_target_sha256,

    "activation_status":
        cell74_activation_status,

    "activation_execution_started":
        cell74_activation_execution_started,

    "activation_committed":
        cell74_activation_committed,

    "rollback_executed":
        cell74_rollback_executed,

    "rollback_verification_pass":
        cell74_rollback_verification_pass,

    "before_bundle_sha256":
        cell67_before_bundle_fingerprint,

    "expected_after_canonical_sha256":
        cell67_after_bundle_fingerprint,

    "expected_after_derived_sha256":
        cell71_after_derived_bundle_sha256,

    "expected_full_after_transaction_sha256":
        cell71_full_after_transaction_sha256,

    "final_live_canonical_sha256":
        cell74_final_live_canonical_sha256,

    "final_live_derived_sha256":
        cell74_final_live_derived_sha256,

    "transaction_exception":
        cell74_transaction_exception,

    "post_activation_failures":
        cell74_clone(
            cell74_post_activation_failures
        ),

    "question_generation_status":
        "NOT_STARTED",

    "expected_answer_generation_status":
        "NOT_STARTED",

    "artifact_write_status":
        "NONE",

    "git_commit_status":
        "NOT_STARTED",

    "git_push_status":
        "NOT_STARTED",

    "pull_request_status":
        "NOT_STARTED",
}


# ============================================================
# 14) Final gates
# ============================================================

cell74_gates = {
    "explicit_user_authorization_is_present":
        (
            CELL74_ACTIVATION_AUTHORIZATION
            ==
            "PROVIDED_BY_USER_IN_CHAT"
        ),

    "v2_exercise_pilot_semantic_adjudication_is_pass":
        (
            CELL74_SEMANTIC_DECISION
            ==
            "PASS"

            and

            cell_73_pass
        ),

    "all_activation_preconditions_were_satisfied":
        (
            len(
                cell74_prerequisite_failures
            )
            ==
            0
        ),

    "live_state_matched_exact_validated_before_immediately_before_activation":
        (
            cell74_live_before_bundle_sha256
            ==
            cell67_before_bundle_fingerprint
        ),

    "activation_execution_started":
        (
            cell74_activation_execution_started
            is True
        ),

    "transaction_completed_without_exception":
        (
            cell74_transaction_exception
            is None
        ),

    "all_post_activation_validations_passed":
        (
            len(
                cell74_post_activation_failures
            )
            ==
            0
        ),

    "rollback_was_not_required":
        (
            cell74_rollback_executed
            is False
        ),

    "activation_is_committed_in_memory":
        (
            cell74_activation_committed
            is True

            and

            cell74_activation_status
            ==
            "ACTIVATED_IN_MEMORY_AND_VALIDATED"
        ),

    "final_live_canonical_state_matches_exact_validated_after_fingerprint":
        (
            cell74_final_live_canonical_sha256
            ==
            cell67_after_bundle_fingerprint
        ),

    "final_live_derived_state_matches_exact_validated_after_fingerprint":
        (
            cell74_final_live_derived_sha256
            ==
            cell71_after_derived_bundle_sha256
        ),

    "rollback_path_is_verified":
        (
            cell74_rollback_verification_pass
            is True
        ),

    "question_generation_remains_not_started":
        (
            EXERCISE_V2_ACTIVATION_EXECUTION_V1[
                "question_generation_status"
            ]
            ==
            "NOT_STARTED"
        ),

    "expected_answer_generation_remains_not_started":
        (
            EXERCISE_V2_ACTIVATION_EXECUTION_V1[
                "expected_answer_generation_status"
            ]
            ==
            "NOT_STARTED"
        ),

    "no_artifact_write":
        (
            EXERCISE_V2_ACTIVATION_EXECUTION_V1[
                "artifact_write_status"
            ]
            ==
            "NONE"
        ),

    "no_commit_push_or_pr":
        (
            EXERCISE_V2_ACTIVATION_EXECUTION_V1[
                "git_commit_status"
            ]
            ==
            "NOT_STARTED"

            and

            EXERCISE_V2_ACTIVATION_EXECUTION_V1[
                "git_push_status"
            ]
            ==
            "NOT_STARTED"

            and

            EXERCISE_V2_ACTIVATION_EXECUTION_V1[
                "pull_request_status"
            ]
            ==
            "NOT_STARTED"
        ),
}


cell_74_pass = all(
    cell74_gates.values()
)


# ============================================================
# 15) Output
# ============================================================

print(
    "=== EXERCISE V2 AUTHORIZED ATOMIC ACTIVATION ==="
)


print(
    "\n=== SEMANTIC DECISION ==="
)


print(
    "Decision:",
    CELL74_SEMANTIC_DECISION
)


print(
    "Evidence SHA256:",
    cell73_joined_target_sha256
)


print(
    "Reason codes:"
)


for reason_code in (
    EXERCISE_V2_PILOT_SEMANTIC_DECISION_V1[
        "reason_codes"
    ]
):

    print(
        "-",
        reason_code
    )


print(
    "\n=== PRECONDITIONS ==="
)


print(
    "Failures:",
    cell74_prerequisite_failures
)


print(
    "Live BEFORE SHA256:",
    cell74_live_before_bundle_sha256
)


print(
    "Expected BEFORE SHA256:",
    cell67_before_bundle_fingerprint
)


print(
    "\n=== TRANSACTION EXECUTION ==="
)


print(
    "Started:",
    cell74_activation_execution_started
)


print(
    "Exception:",
    cell74_transaction_exception
)


print(
    "Post-activation failures:",
    cell74_post_activation_failures
)


print(
    "Rollback executed:",
    cell74_rollback_executed
)


print(
    "Rollback verification:",
    cell74_rollback_verification_pass
)


print(
    "Activation committed:",
    cell74_activation_committed
)


print(
    "\nACTIVATION STATUS:",
    cell74_activation_status
)


print(
    "\n=== FINAL FINGERPRINTS ==="
)


print(
    "Final live canonical:",
    cell74_final_live_canonical_sha256
)


print(
    "Expected AFTER canonical:",
    cell67_after_bundle_fingerprint
)


print(
    "Final live derived:",
    cell74_final_live_derived_sha256
)


print(
    "Expected AFTER derived:",
    cell71_after_derived_bundle_sha256
)


print(
    "\n=== FINAL ACTIVE COUNTS ==="
)


print(
    "Golden:",
    len(
        golden_case_records
    )
)


print(
    "Auto:",
    len(
        auto_case_records
    )
)


print(
    "Groundability:",
    len(
        case_groundability_precheck
    )
)


print(
    "Task Design:",
    len(
        case_task_design
    )
)


print(
    "Tasks:",
    len(
        task_draft_plan_rows
    )
)


print(
    "Targets:",
    len(
        target_evidence_records
    )
)


print(
    "Pilots:",
    len(
        pilot_selection_rows
    )
)


print(
    "Question packets:",
    len(
        question_drafting_pilot_packets
    )
)


print(
    "\n=== ACTIVE EXERCISE PILOT ==="
)


cell74_final_active_exercise_selection = [
    row

    for row
    in pilot_selection_rows

    if (
        row[
            "source_type"
        ]
        ==
        "book"

        and

        row[
            "stratum"
        ]
        ==
        "exercise_or_self_assessment"
    )
]


cell74_final_active_exercise_packets = [
    packet

    for packet
    in question_drafting_pilot_packets

    if (
        packet[
            "source_type"
        ]
        ==
        "book"

        and

        packet[
            "stratum"
        ]
        ==
        "exercise_or_self_assessment"
    )
]


print(
    "Selection:",
    (
        cell74_final_active_exercise_selection[
            0
        ]
        if (
            len(
                cell74_final_active_exercise_selection
            )
            ==
            1
        )
        else cell74_final_active_exercise_selection
    )
)


print(
    "Question packet ID:",
    (
        cell74_final_active_exercise_packets[
            0
        ][
            "pilot_case_id"
        ]
        if (
            len(
                cell74_final_active_exercise_packets
            )
            ==
            1
        )
        else None
    )
)


print(
    "\n=== CODE CELL 74 GATES ==="
)


for gate_name, passed in (
    cell74_gates.items()
):

    print(
        gate_name,
        ":",
        "PASS"
        if passed
        else "FAIL"
    )


print(
    "\nFAILED GATES:",
    [
        gate_name

        for gate_name, passed
        in cell74_gates.items()

        if not passed
    ]
)


print(
    "\nCODE CELL 74 GATE:",
    "PASS"
    if cell_74_pass
    else "FAIL"
)


assert cell_74_pass, (
    "Exercise V2 activation did not complete successfully. "
    "If the transaction began, automatic rollback has already "
    "been attempted. Do not continue to question generation. "
    "Repair this SAME Code Cell 74 using the printed "
    "precondition, transaction, rollback and fingerprint "
    "diagnostics."
)
```

```python
# Code Cell 75 — REPAIRED

# ============================================================
# Post-Activation Closure and Active-State Freeze
#
# REPAIR:
#   Do NOT assume the Groundability precheck field name.
#   Discover the exact field empirically from the ACTIVE
#   Exercise V2 records by its already-proven semantic value:
#
#       EVIDENCE_RECOVERABILITY_PASS
#
# READ ONLY.
#
# NO activation.
# NO mutation.
# NO question generation.
# NO artifact write.
# ============================================================


# ============================================================
# 1) Helpers
# ============================================================

def cell75_sha(
    value,
):

    return hashlib.sha256(
        stable_json_key(
            value
        ).encode(
            "utf-8"
        )
    ).hexdigest()


def cell75_clone(
    value,
):

    return cell52_safe_clone(
        value
    )


def cell75_field_counter(
    rows,
    field_name,
):

    return Counter(
        row.get(
            field_name,
            "__MISSING_FIELD__",
        )

        for row
        in rows
    )


def cell75_uniform_value_field_candidates(
    rows,
    expected_value,
):

    if not rows:
        return []


    common_keys = set(
        rows[
            0
        ].keys()
    )


    for row in (
        rows[
            1:
        ]
    ):

        common_keys &= set(
            row.keys()
        )


    return sorted(
        [
            field_name

            for field_name
            in common_keys

            if all(
                row.get(
                    field_name
                )
                ==
                expected_value

                for row
                in rows
            )
        ]
    )


# ============================================================
# 2) Preconditions
# ============================================================

cell75_prerequisite_failures = []


if not (
    cell_74_pass
):

    cell75_prerequisite_failures.append(
        "CELL74_NOT_PASS"
    )


if (
    cell74_activation_status
    !=
    "ACTIVATED_IN_MEMORY_AND_VALIDATED"
):

    cell75_prerequisite_failures.append(
        {
            "reason":
                "CELL74_ACTIVATION_STATUS_UNEXPECTED",

            "actual":
                cell74_activation_status,
        }
    )


if not (
    cell74_activation_committed
):

    cell75_prerequisite_failures.append(
        "CELL74_ACTIVATION_NOT_COMMITTED"
    )


if (
    cell74_rollback_executed
):

    cell75_prerequisite_failures.append(
        "UNEXPECTED_CELL74_ROLLBACK"
    )


# ============================================================
# 3) Snapshot ACTIVE state before closure
# ============================================================

cell75_active_registry_payload = {
    "golden_case_records":
        golden_case_records,

    "auto_case_records":
        auto_case_records,

    "case_groundability_precheck":
        case_groundability_precheck,

    "case_task_design":
        case_task_design,

    "task_draft_plan_rows":
        task_draft_plan_rows,

    "target_evidence_records":
        target_evidence_records,

    "pilot_selection_rows":
        pilot_selection_rows,
}


cell75_active_index_payload = {
    "golden_case_records_by_id":
        golden_case_records_by_id,

    "task_draft_plans":
        task_draft_plans,

    "target_evidence_by_case_id":
        target_evidence_by_case_id,

    "target_evidence_by_task_draft_id":
        target_evidence_by_task_draft_id,
}


cell75_active_derived_payload = {
    "candidate_hash_to_case_ids":
        candidate_hash_to_case_ids,

    "cross_stratum_repeated_candidate_hashes":
        cross_stratum_repeated_candidate_hashes,

    "question_drafting_pilot_packets":
        question_drafting_pilot_packets,

    "question_drafting_pilot_by_id":
        question_drafting_pilot_by_id,

    "target_id_counts":
        target_id_counts,
}


cell75_active_canonical_sha256 = (
    cell75_sha(
        {
            "registries":
                cell75_active_registry_payload,

            "indexes":
                cell75_active_index_payload,
        }
    )
)


cell75_active_derived_sha256 = (
    cell75_sha(
        cell75_active_derived_payload
    )
)


cell75_live_state_snapshot_before = {
    "canonical":
        cell75_active_canonical_sha256,

    "derived":
        cell75_active_derived_sha256,
}


# ============================================================
# 4) Exact fingerprint closure
# ============================================================

cell75_fingerprint_failures = []


if (
    cell75_active_canonical_sha256
    !=
    cell67_after_bundle_fingerprint
):

    cell75_fingerprint_failures.append(
        {
            "reason":
                "ACTIVE_CANONICAL_NOT_EQUAL_VALIDATED_AFTER",

            "expected":
                cell67_after_bundle_fingerprint,

            "actual":
                cell75_active_canonical_sha256,
        }
    )


if (
    cell75_active_derived_sha256
    !=
    cell71_after_derived_bundle_sha256
):

    cell75_fingerprint_failures.append(
        {
            "reason":
                "ACTIVE_DERIVED_NOT_EQUAL_VALIDATED_AFTER",

            "expected":
                cell71_after_derived_bundle_sha256,

            "actual":
                cell75_active_derived_sha256,
        }
    )


# ============================================================
# 5) Resolve ACTIVE Exercise V2 slice
# ============================================================

CELL75_EXERCISE_SOURCE_TYPE = (
    "book"
)


CELL75_EXERCISE_STRATUM = (
    "exercise_or_self_assessment"
)


cell75_active_exercise_cases = [
    row

    for row
    in golden_case_records

    if (
        row.get(
            "source_type"
        )
        ==
        CELL75_EXERCISE_SOURCE_TYPE

        and

        row.get(
            "stratum"
        )
        ==
        CELL75_EXERCISE_STRATUM
    )
]


cell75_active_exercise_case_ids = {
    row[
        "golden_case_id"
    ]

    for row
    in cell75_active_exercise_cases
}


cell75_active_exercise_groundability = [
    row

    for row
    in case_groundability_precheck

    if (
        row.get(
            "golden_case_id"
        )
        in
        cell75_active_exercise_case_ids
    )
]


cell75_active_exercise_task_design = [
    row

    for row
    in case_task_design

    if (
        row.get(
            "golden_case_id"
        )
        in
        cell75_active_exercise_case_ids
    )
]


cell75_active_exercise_tasks = [
    row

    for row
    in task_draft_plan_rows

    if (
        row.get(
            "golden_case_id"
        )
        in
        cell75_active_exercise_case_ids
    )
]


cell75_active_exercise_task_ids = {
    row[
        "task_draft_id"
    ]

    for row
    in cell75_active_exercise_tasks
}


cell75_active_exercise_targets = [
    row

    for row
    in target_evidence_records

    if (
        row.get(
            "golden_case_id"
        )
        in
        cell75_active_exercise_case_ids
    )
]


cell75_active_exercise_target_ids = {
    row[
        "target_evidence_id"
    ]

    for row
    in cell75_active_exercise_targets
}


cell75_active_exercise_pilot_selections = [
    row

    for row
    in pilot_selection_rows

    if (
        row.get(
            "source_type"
        )
        ==
        CELL75_EXERCISE_SOURCE_TYPE

        and

        row.get(
            "stratum"
        )
        ==
        CELL75_EXERCISE_STRATUM
    )
]


cell75_active_exercise_question_packets = [
    packet

    for packet
    in question_drafting_pilot_packets

    if (
        packet.get(
            "source_type"
        )
        ==
        CELL75_EXERCISE_SOURCE_TYPE

        and

        packet.get(
            "stratum"
        )
        ==
        CELL75_EXERCISE_STRATUM
    )
]


# ============================================================
# 6) Slice cardinality
# ============================================================

cell75_slice_counts = {
    "cases":
        len(
            cell75_active_exercise_cases
        ),

    "groundability":
        len(
            cell75_active_exercise_groundability
        ),

    "task_design":
        len(
            cell75_active_exercise_task_design
        ),

    "tasks":
        len(
            cell75_active_exercise_tasks
        ),

    "targets":
        len(
            cell75_active_exercise_targets
        ),

    "pilot_selections":
        len(
            cell75_active_exercise_pilot_selections
        ),

    "question_packets":
        len(
            cell75_active_exercise_question_packets
        ),
}


CELL75_EXPECTED_SLICE_COUNTS = {
    "cases":
        38,

    "groundability":
        38,

    "task_design":
        38,

    "tasks":
        38,

    "targets":
        38,

    "pilot_selections":
        1,

    "question_packets":
        1,
}


cell75_slice_count_failures = [
    {
        "object":
            object_name,

        "expected":
            expected_count,

        "actual":
            cell75_slice_counts[
                object_name
            ],
    }

    for object_name, expected_count
    in CELL75_EXPECTED_SLICE_COUNTS.items()

    if (
        cell75_slice_counts[
            object_name
        ]
        !=
        expected_count
    )
]


# ============================================================
# 7) Exact V2 identity sets
# ============================================================

cell75_v2_identity_failures = []


cell75_expected_v2_case_ids = set(
    cell65_v2_case_ids
)


cell75_expected_v2_task_ids = set(
    cell65_v2_task_ids
)


cell75_expected_v2_target_ids = set(
    cell65_v2_target_ids
)


if (
    cell75_active_exercise_case_ids
    !=
    cell75_expected_v2_case_ids
):

    cell75_v2_identity_failures.append(
        {
            "reason":
                "ACTIVE_CASE_SET_NOT_EXACT_V2",

            "missing":
                sorted(
                    cell75_expected_v2_case_ids
                    -
                    cell75_active_exercise_case_ids
                ),

            "unexpected":
                sorted(
                    cell75_active_exercise_case_ids
                    -
                    cell75_expected_v2_case_ids
                ),
        }
    )


if (
    cell75_active_exercise_task_ids
    !=
    cell75_expected_v2_task_ids
):

    cell75_v2_identity_failures.append(
        {
            "reason":
                "ACTIVE_TASK_SET_NOT_EXACT_V2",

            "missing":
                sorted(
                    cell75_expected_v2_task_ids
                    -
                    cell75_active_exercise_task_ids
                ),

            "unexpected":
                sorted(
                    cell75_active_exercise_task_ids
                    -
                    cell75_expected_v2_task_ids
                ),
        }
    )


if (
    cell75_active_exercise_target_ids
    !=
    cell75_expected_v2_target_ids
):

    cell75_v2_identity_failures.append(
        {
            "reason":
                "ACTIVE_TARGET_SET_NOT_EXACT_V2",

            "missing":
                sorted(
                    cell75_expected_v2_target_ids
                    -
                    cell75_active_exercise_target_ids
                ),

            "unexpected":
                sorted(
                    cell75_active_exercise_target_ids
                    -
                    cell75_expected_v2_target_ids
                ),
        }
    )


# ============================================================
# 8) Task lifecycle
# ============================================================

cell75_task_construction_counts = (
    cell75_field_counter(
        cell75_active_exercise_tasks,
        "construction_eligibility",
    )
)


cell75_task_target_status_counts = (
    cell75_field_counter(
        cell75_active_exercise_tasks,
        "target_evidence_status",
    )
)


cell75_task_question_status_counts = (
    cell75_field_counter(
        cell75_active_exercise_tasks,
        "question_text_status",
    )
)


cell75_task_expected_answer_status_counts = (
    cell75_field_counter(
        cell75_active_exercise_tasks,
        "expected_answer_status",
    )
)


cell75_task_citation_status_counts = (
    cell75_field_counter(
        cell75_active_exercise_tasks,
        "citation_target_status",
    )
)


cell75_task_lifecycle_failures = []


CELL75_EXPECTED_TASK_LIFECYCLE = {
    "construction_eligibility":
        Counter(
            {
                "READY_FOR_QUESTION_DRAFT":
                    38
            }
        ),

    "target_evidence_status":
        Counter(
            {
                "SELECTED_AND_VALIDATED":
                    38
            }
        ),

    "question_text_status":
        Counter(
            {
                "NOT_STARTED":
                    38
            }
        ),

    "expected_answer_status":
        Counter(
            {
                "NOT_STARTED":
                    38
            }
        ),

    "citation_target_status":
        Counter(
            {
                "NOT_STARTED":
                    38
            }
        ),
}


cell75_actual_task_lifecycle = {
    "construction_eligibility":
        cell75_task_construction_counts,

    "target_evidence_status":
        cell75_task_target_status_counts,

    "question_text_status":
        cell75_task_question_status_counts,

    "expected_answer_status":
        cell75_task_expected_answer_status_counts,

    "citation_target_status":
        cell75_task_citation_status_counts,
}


for field_name, expected_counter in (
    CELL75_EXPECTED_TASK_LIFECYCLE.items()
):

    if (
        cell75_actual_task_lifecycle[
            field_name
        ]
        !=
        expected_counter
    ):

        cell75_task_lifecycle_failures.append(
            {
                "field":
                    field_name,

                "expected":
                    dict(
                        expected_counter
                    ),

                "actual":
                    dict(
                        cell75_actual_task_lifecycle[
                            field_name
                        ]
                    ),
            }
        )


# ============================================================
# 9) Groundability lifecycle
#
# IMPORTANT REPAIR:
#
# The semantic VALUE is known from Cell62:
#
#   EVIDENCE_RECOVERABILITY_PASS
#
# But the FIELD NAME is discovered from the real ACTIVE
# Groundability records instead of being guessed.
# ============================================================

cell75_groundability_construction_counts = (
    cell75_field_counter(
        cell75_active_exercise_groundability,
        "construction_eligibility",
    )
)


cell75_groundability_question_counts = (
    cell75_field_counter(
        cell75_active_exercise_groundability,
        "question_generation_status",
    )
)


cell75_groundability_answer_counts = (
    cell75_field_counter(
        cell75_active_exercise_groundability,
        "expected_answer_generation_status",
    )
)


cell75_groundability_precheck_field_candidates = (
    cell75_uniform_value_field_candidates(
        cell75_active_exercise_groundability,
        "EVIDENCE_RECOVERABILITY_PASS",
    )
)


cell75_groundability_precheck_field = (
    cell75_groundability_precheck_field_candidates[
        0
    ]

    if (
        len(
            cell75_groundability_precheck_field_candidates
        )
        ==
        1
    )

    else None
)


cell75_groundability_precheck_counts = (
    cell75_field_counter(
        cell75_active_exercise_groundability,
        cell75_groundability_precheck_field,
    )

    if (
        cell75_groundability_precheck_field
        is not None
    )

    else Counter()
)


cell75_groundability_lifecycle_failures = []


if (
    len(
        cell75_groundability_precheck_field_candidates
    )
    !=
    1
):

    cell75_groundability_lifecycle_failures.append(
        {
            "reason":
                "GROUNDABILITY_PRECHECK_FIELD_NOT_UNIQUELY_DISCOVERED",

            "candidates":
                cell75_groundability_precheck_field_candidates,

            "sample_keys":
                (
                    sorted(
                        cell75_active_exercise_groundability[
                            0
                        ].keys()
                    )
                    if (
                        cell75_active_exercise_groundability
                    )
                    else []
                ),
        }
    )


if (
    cell75_groundability_construction_counts
    !=
    Counter(
        {
            "READY_FOR_GROUNDED_TASK_DRAFTING":
                38
        }
    )
):

    cell75_groundability_lifecycle_failures.append(
        {
            "field":
                "construction_eligibility",

            "actual":
                dict(
                    cell75_groundability_construction_counts
                ),
        }
    )


if (
    cell75_groundability_precheck_field
    is not None

    and

    cell75_groundability_precheck_counts
    !=
    Counter(
        {
            "EVIDENCE_RECOVERABILITY_PASS":
                38
        }
    )
):

    cell75_groundability_lifecycle_failures.append(
        {
            "field":
                cell75_groundability_precheck_field,

            "actual":
                dict(
                    cell75_groundability_precheck_counts
                ),
        }
    )


if (
    cell75_groundability_question_counts
    !=
    Counter(
        {
            "NOT_STARTED":
                38
        }
    )
):

    cell75_groundability_lifecycle_failures.append(
        {
            "field":
                "question_generation_status",

            "actual":
                dict(
                    cell75_groundability_question_counts
                ),
        }
    )


if (
    cell75_groundability_answer_counts
    !=
    Counter(
        {
            "NOT_STARTED":
                38
        }
    )
):

    cell75_groundability_lifecycle_failures.append(
        {
            "field":
                "expected_answer_generation_status",

            "actual":
                dict(
                    cell75_groundability_answer_counts
                ),
        }
    )


# ============================================================
# 10) Task Design lifecycle
# ============================================================

cell75_task_design_construction_counts = (
    cell75_field_counter(
        cell75_active_exercise_task_design,
        "task_construction_status",
    )
)


cell75_task_design_question_counts = (
    cell75_field_counter(
        cell75_active_exercise_task_design,
        "question_generation_status",
    )
)


cell75_task_design_answer_counts = (
    cell75_field_counter(
        cell75_active_exercise_task_design,
        "expected_answer_generation_status",
    )
)


cell75_task_design_lifecycle_failures = []


if (
    cell75_task_design_construction_counts
    !=
    Counter(
        {
            "READY_FOR_TASK_CONSTRUCTION":
                38
        }
    )
):

    cell75_task_design_lifecycle_failures.append(
        {
            "field":
                "task_construction_status",

            "actual":
                dict(
                    cell75_task_design_construction_counts
                ),
        }
    )


if (
    cell75_task_design_question_counts
    !=
    Counter(
        {
            "NOT_STARTED":
                38
        }
    )
):

    cell75_task_design_lifecycle_failures.append(
        {
            "field":
                "question_generation_status",

            "actual":
                dict(
                    cell75_task_design_question_counts
                ),
        }
    )


if (
    cell75_task_design_answer_counts
    !=
    Counter(
        {
            "NOT_STARTED":
                38
        }
    )
):

    cell75_task_design_lifecycle_failures.append(
        {
            "field":
                "expected_answer_generation_status",

            "actual":
                dict(
                    cell75_task_design_answer_counts
                ),
        }
    )


# ============================================================
# 11) Target lifecycle / V2 semantics
# ============================================================

cell75_target_method_counts = (
    cell75_field_counter(
        cell75_active_exercise_targets,
        "evidence_target_method",
    )
)


cell75_target_basis_counts = (
    cell75_field_counter(
        cell75_active_exercise_targets,
        "target_selection_basis",
    )
)


cell75_target_question_counts = (
    cell75_field_counter(
        cell75_active_exercise_targets,
        "question_generation_status",
    )
)


cell75_target_answer_counts = (
    cell75_field_counter(
        cell75_active_exercise_targets,
        "expected_answer_generation_status",
    )
)


cell75_target_lifecycle_failures = []


if (
    cell75_target_method_counts
    !=
    Counter(
        {
            "USE_EXPLICIT_BOOK_EXERCISE_UNIT":
                38
        }
    )
):

    cell75_target_lifecycle_failures.append(
        {
            "field":
                "evidence_target_method",

            "actual":
                dict(
                    cell75_target_method_counts
                ),
        }
    )


if (
    cell75_target_basis_counts
    !=
    Counter(
        {
            "EXACT_ADJUDICATED_EXERCISE_UNIT":
                38
        }
    )
):

    cell75_target_lifecycle_failures.append(
        {
            "field":
                "target_selection_basis",

            "actual":
                dict(
                    cell75_target_basis_counts
                ),
        }
    )


if (
    cell75_target_question_counts
    !=
    Counter(
        {
            "NOT_STARTED":
                38
        }
    )
):

    cell75_target_lifecycle_failures.append(
        {
            "field":
                "question_generation_status",

            "actual":
                dict(
                    cell75_target_question_counts
                ),
        }
    )


if (
    cell75_target_answer_counts
    !=
    Counter(
        {
            "NOT_STARTED":
                38
        }
    )
):

    cell75_target_lifecycle_failures.append(
        {
            "field":
                "expected_answer_generation_status",

            "actual":
                dict(
                    cell75_target_answer_counts
                ),
        }
    )


# ============================================================
# 12) Exact Active Exercise Pilot
# ============================================================

cell75_pilot_failures = []


cell75_active_exercise_pilot_selection = (
    cell75_active_exercise_pilot_selections[
        0
    ]

    if (
        len(
            cell75_active_exercise_pilot_selections
        )
        ==
        1
    )

    else None
)


cell75_active_exercise_question_packet = (
    cell75_active_exercise_question_packets[
        0
    ]

    if (
        len(
            cell75_active_exercise_question_packets
        )
        ==
        1
    )

    else None
)


if (
    cell75_active_exercise_pilot_selection
    is None
):

    cell75_pilot_failures.append(
        "ACTIVE_EXERCISE_PILOT_SELECTION_NOT_UNIQUE"
    )


else:

    for field_name in (
        [
            "golden_case_id",
            "task_draft_id",
            "target_evidence_id",
        ]
    ):

        if (
            cell75_active_exercise_pilot_selection.get(
                field_name
            )
            !=
            staged_exercise_v2_pilot_replacement_row[
                field_name
            ]
        ):

            cell75_pilot_failures.append(
                {
                    "object":
                        "pilot_selection",

                    "field":
                        field_name,

                    "expected":
                        staged_exercise_v2_pilot_replacement_row[
                            field_name
                        ],

                    "actual":
                        cell75_active_exercise_pilot_selection.get(
                            field_name
                        ),
                }
            )


if (
    cell75_active_exercise_question_packet
    is None
):

    cell75_pilot_failures.append(
        "ACTIVE_EXERCISE_QUESTION_PACKET_NOT_UNIQUE"
    )


else:

    if (
        cell75_active_exercise_question_packet.get(
            "pilot_case_id"
        )
        !=
        cell71_v2_exercise_packet_id
    ):

        cell75_pilot_failures.append(
            {
                "field":
                    "pilot_case_id",

                "expected":
                    cell71_v2_exercise_packet_id,

                "actual":
                    cell75_active_exercise_question_packet.get(
                        "pilot_case_id"
                    ),
            }
        )


    for field_name in (
        [
            "golden_case_id",
            "task_draft_id",
            "target_evidence_id",
        ]
    ):

        if (
            cell75_active_exercise_question_packet.get(
                field_name
            )
            !=
            staged_exercise_v2_pilot_replacement_row[
                field_name
            ]
        ):

            cell75_pilot_failures.append(
                {
                    "object":
                        "question_packet",

                    "field":
                        field_name,

                    "expected":
                        staged_exercise_v2_pilot_replacement_row[
                            field_name
                        ],

                    "actual":
                        cell75_active_exercise_question_packet.get(
                            field_name
                        ),
                }
            )


# ============================================================
# 13) Current-State Legacy leakage
# ============================================================

cell75_active_current_state_serialized = (
    stable_json_key(
        {
            "canonical_registries":
                cell75_active_registry_payload,

            "canonical_indexes":
                cell75_active_index_payload,

            "derived":
                cell75_active_derived_payload,
        }
    )
)


cell75_active_legacy_leakage = {
    "case_ids":
        sorted(
            [
                item_id

                for item_id
                in cell65_legacy_exercise_case_ids

                if (
                    item_id
                    in
                    cell75_active_current_state_serialized
                )
            ]
        ),

    "task_ids":
        sorted(
            [
                item_id

                for item_id
                in cell65_legacy_exercise_task_ids

                if (
                    item_id
                    in
                    cell75_active_current_state_serialized
                )
            ]
        ),

    "target_ids":
        sorted(
            [
                item_id

                for item_id
                in cell65_legacy_exercise_target_ids

                if (
                    item_id
                    in
                    cell75_active_current_state_serialized
                )
            ]
        ),
}


# ============================================================
# 14) Historical / SAFE isolation
# ============================================================

cell75_historical_snapshot_now = {
    variable_name:
        cell75_sha(
            globals()[
                variable_name
            ]
        )

    for variable_name
    in cell69_v2_historical_preserve_objects
}


cell75_safe_snapshot_now = {
    variable_name:
        cell75_sha(
            globals()[
                variable_name
            ]
        )

    for variable_name
    in cell69_v2_safe_no_action_objects
}


cell75_historical_isolation_failures = []


if (
    cell75_historical_snapshot_now
    !=
    cell74_historical_snapshot_before
):

    cell75_historical_isolation_failures.append(
        "HISTORICAL_PRESERVE_OBJECTS_CHANGED"
    )


if (
    cell75_safe_snapshot_now
    !=
    cell74_safe_snapshot_before
):

    cell75_historical_isolation_failures.append(
        "SAFE_NO_ACTION_OBJECTS_CHANGED"
    )


# ============================================================
# 15) Active Exercise packet generation state
# ============================================================

cell75_active_packet_generation_state = {
    field_name:
        (
            cell75_active_exercise_question_packet.get(
                field_name
            )

            if (
                cell75_active_exercise_question_packet
                is not None
            )

            else None
        )

    for field_name
    in [
        "question_draft_status",
        "question_groundability_status",
        "expected_answer_status",
        "citation_target_validation_status",
    ]
}


cell75_packet_generation_state_failures = [
    {
        "field":
            field_name,

        "actual":
            value,
    }

    for field_name, value
    in cell75_active_packet_generation_state.items()

    if (
        value
        !=
        "NOT_STARTED"
    )
]


# ============================================================
# 16) Aggregate closure failures
# ============================================================

cell75_closure_failures = []


for failure_family, failures in (
    [
        (
            "prerequisite",
            cell75_prerequisite_failures,
        ),
        (
            "fingerprint",
            cell75_fingerprint_failures,
        ),
        (
            "slice_count",
            cell75_slice_count_failures,
        ),
        (
            "v2_identity",
            cell75_v2_identity_failures,
        ),
        (
            "task_lifecycle",
            cell75_task_lifecycle_failures,
        ),
        (
            "groundability_lifecycle",
            cell75_groundability_lifecycle_failures,
        ),
        (
            "task_design_lifecycle",
            cell75_task_design_lifecycle_failures,
        ),
        (
            "target_lifecycle",
            cell75_target_lifecycle_failures,
        ),
        (
            "pilot",
            cell75_pilot_failures,
        ),
        (
            "historical_isolation",
            cell75_historical_isolation_failures,
        ),
        (
            "packet_generation_state",
            cell75_packet_generation_state_failures,
        ),
    ]
):

    if failures:

        cell75_closure_failures.append(
            {
                "family":
                    failure_family,

                "detail":
                    failures,
            }
        )


if (
    any(
        cell75_active_legacy_leakage.values()
    )
):

    cell75_closure_failures.append(
        {
            "family":
                "active_legacy_leakage",

            "detail":
                cell75_active_legacy_leakage,
        }
    )


# ============================================================
# 17) Final read-only stability check
# ============================================================

cell75_final_canonical_sha256 = (
    cell75_sha(
        {
            "registries":
                {
                    "golden_case_records":
                        golden_case_records,

                    "auto_case_records":
                        auto_case_records,

                    "case_groundability_precheck":
                        case_groundability_precheck,

                    "case_task_design":
                        case_task_design,

                    "task_draft_plan_rows":
                        task_draft_plan_rows,

                    "target_evidence_records":
                        target_evidence_records,

                    "pilot_selection_rows":
                        pilot_selection_rows,
                },

            "indexes":
                {
                    "golden_case_records_by_id":
                        golden_case_records_by_id,

                    "task_draft_plans":
                        task_draft_plans,

                    "target_evidence_by_case_id":
                        target_evidence_by_case_id,

                    "target_evidence_by_task_draft_id":
                        target_evidence_by_task_draft_id,
                },
        }
    )
)


cell75_final_derived_sha256 = (
    cell75_sha(
        {
            "candidate_hash_to_case_ids":
                candidate_hash_to_case_ids,

            "cross_stratum_repeated_candidate_hashes":
                cross_stratum_repeated_candidate_hashes,

            "question_drafting_pilot_packets":
                question_drafting_pilot_packets,

            "question_drafting_pilot_by_id":
                question_drafting_pilot_by_id,

            "target_id_counts":
                target_id_counts,
        }
    )
)


cell75_live_state_snapshot_after = {
    "canonical":
        cell75_final_canonical_sha256,

    "derived":
        cell75_final_derived_sha256,
}


cell75_readonly_stability_failure = (
    None

    if (
        cell75_live_state_snapshot_before
        ==
        cell75_live_state_snapshot_after
    )

    else
    {
        "before":
            cell75_live_state_snapshot_before,

        "after":
            cell75_live_state_snapshot_after,
    }
)


if (
    cell75_readonly_stability_failure
    is not None
):

    cell75_closure_failures.append(
        {
            "family":
                "readonly_stability",

            "detail":
                cell75_readonly_stability_failure,
        }
    )


# ============================================================
# 18) Freeze post-activation closure
# ============================================================

cell75_post_activation_closed = (
    len(
        cell75_closure_failures
    )
    ==
    0
)


cell75_post_activation_status = (
    "EXERCISE_V2_ACTIVATION_CLOSED_"
    "READY_FOR_QUESTION_DRAFTING"

    if (
        cell75_post_activation_closed
    )

    else
    "EXERCISE_V2_POST_ACTIVATION_CLOSURE_FAILED"
)


EXERCISE_V2_POST_ACTIVATION_CLOSURE_V1 = {
    "version":
        "exercise_v2_post_activation_closure_v1",

    "status":
        cell75_post_activation_status,

    "activation_execution_status":
        cell74_activation_status,

    "active_canonical_sha256":
        cell75_active_canonical_sha256,

    "active_derived_sha256":
        cell75_active_derived_sha256,

    "exercise_slice_counts":
        cell75_clone(
            cell75_slice_counts
        ),

    "groundability_precheck_field_discovered":
        cell75_groundability_precheck_field,

    "groundability_precheck_field_candidates":
        list(
            cell75_groundability_precheck_field_candidates
        ),

    "task_construction_counts":
        dict(
            cell75_task_construction_counts
        ),

    "task_target_status_counts":
        dict(
            cell75_task_target_status_counts
        ),

    "groundability_construction_counts":
        dict(
            cell75_groundability_construction_counts
        ),

    "groundability_precheck_counts":
        dict(
            cell75_groundability_precheck_counts
        ),

    "target_method_counts":
        dict(
            cell75_target_method_counts
        ),

    "target_basis_counts":
        dict(
            cell75_target_basis_counts
        ),

    "active_exercise_pilot": {
        "golden_case_id":
            (
                cell75_active_exercise_pilot_selection.get(
                    "golden_case_id"
                )

                if (
                    cell75_active_exercise_pilot_selection
                    is not None
                )

                else None
            ),

        "task_draft_id":
            (
                cell75_active_exercise_pilot_selection.get(
                    "task_draft_id"
                )

                if (
                    cell75_active_exercise_pilot_selection
                    is not None
                )

                else None
            ),

        "target_evidence_id":
            (
                cell75_active_exercise_pilot_selection.get(
                    "target_evidence_id"
                )

                if (
                    cell75_active_exercise_pilot_selection
                    is not None
                )

                else None
            ),

        "question_pilot_id":
            (
                cell75_active_exercise_question_packet.get(
                    "pilot_case_id"
                )

                if (
                    cell75_active_exercise_question_packet
                    is not None
                )

                else None
            ),
    },

    "semantic_decision":
        CELL74_SEMANTIC_DECISION,

    "semantic_evidence_sha256":
        cell73_joined_target_sha256,

    "legacy_leakage":
        cell75_clone(
            cell75_active_legacy_leakage
        ),

    "historical_preserve_policy":
        "UNCHANGED_AND_SEPARATE_FROM_CURRENT_STATE",

    "question_generation_status":
        "NOT_STARTED",

    "expected_answer_generation_status":
        "NOT_STARTED",

    "artifact_write_status":
        "NONE",

    "git_commit_status":
        "NOT_STARTED",

    "git_push_status":
        "NOT_STARTED",

    "pull_request_status":
        "NOT_STARTED",
}


# ============================================================
# 19) Gates
# ============================================================

cell75_gates = {
    "cell74_activation_is_committed_and_validated":
        (
            len(
                cell75_prerequisite_failures
            )
            ==
            0
        ),

    "active_canonical_fingerprint_equals_validated_after":
        (
            cell75_active_canonical_sha256
            ==
            cell67_after_bundle_fingerprint

            and

            len(
                cell75_fingerprint_failures
            )
            ==
            0
        ),

    "active_derived_fingerprint_equals_validated_after":
        (
            cell75_active_derived_sha256
            ==
            cell71_after_derived_bundle_sha256
        ),

    "active_exercise_slice_has_exact_38_38_38_38_38_1_1_cardinality":
        (
            len(
                cell75_slice_count_failures
            )
            ==
            0
        ),

    "active_case_task_target_identity_sets_are_exact_v2_sets":
        (
            len(
                cell75_v2_identity_failures
            )
            ==
            0
        ),

    "all_38_tasks_are_ready_for_question_drafting":
        (
            len(
                cell75_task_lifecycle_failures
            )
            ==
            0
        ),

    "groundability_precheck_field_is_discovered_from_real_active_schema":
        (
            cell75_groundability_precheck_field
            is not None

            and

            len(
                cell75_groundability_precheck_field_candidates
            )
            ==
            1
        ),

    "all_38_groundability_records_are_valid_and_pregeneration":
        (
            len(
                cell75_groundability_lifecycle_failures
            )
            ==
            0
        ),

    "all_38_task_design_records_are_pregeneration":
        (
            len(
                cell75_task_design_lifecycle_failures
            )
            ==
            0
        ),

    "all_38_targets_use_exact_v2_exercise_unit_semantics":
        (
            len(
                cell75_target_lifecycle_failures
            )
            ==
            0
        ),

    "active_exercise_pilot_and_packet_are_exact_semantically_approved_v2_chain":
        (
            len(
                cell75_pilot_failures
            )
            ==
            0
        ),

    "no_legacy_exercise_case_task_or_target_exists_in_current_active_state":
        (
            not any(
                cell75_active_legacy_leakage.values()
            )
        ),

    "historical_and_safe_objects_remain_unchanged":
        (
            len(
                cell75_historical_isolation_failures
            )
            ==
            0
        ),

    "question_answer_and_citation_generation_remain_not_started":
        (
            len(
                cell75_packet_generation_state_failures
            )
            ==
            0
        ),

    "closure_cell_is_read_only_against_active_state":
        (
            cell75_readonly_stability_failure
            is None
        ),

    "post_activation_closure_has_zero_failures":
        (
            len(
                cell75_closure_failures
            )
            ==
            0
        ),

    "exercise_v2_activation_is_closed_ready_for_question_drafting":
        (
            cell75_post_activation_status
            ==
            "EXERCISE_V2_ACTIVATION_CLOSED_"
            "READY_FOR_QUESTION_DRAFTING"
        ),

    "no_artifact_write":
        (
            EXERCISE_V2_POST_ACTIVATION_CLOSURE_V1[
                "artifact_write_status"
            ]
            ==
            "NONE"
        ),

    "no_commit_push_or_pr":
        (
            EXERCISE_V2_POST_ACTIVATION_CLOSURE_V1[
                "git_commit_status"
            ]
            ==
            "NOT_STARTED"

            and

            EXERCISE_V2_POST_ACTIVATION_CLOSURE_V1[
                "git_push_status"
            ]
            ==
            "NOT_STARTED"

            and

            EXERCISE_V2_POST_ACTIVATION_CLOSURE_V1[
                "pull_request_status"
            ]
            ==
            "NOT_STARTED"
        ),
}


cell_75_pass = all(
    cell75_gates.values()
)


# ============================================================
# 20) Output
# ============================================================

print(
    "=== EXERCISE V2 POST-ACTIVATION CLOSURE — REPAIRED ==="
)


print(
    "\n=== ACTIVATION ==="
)


print(
    "Cell74 status:",
    cell74_activation_status
)


print(
    "Prerequisite failures:",
    cell75_prerequisite_failures
)


print(
    "\n=== ACTIVE FINGERPRINTS ==="
)


print(
    "Canonical:",
    cell75_active_canonical_sha256
)


print(
    "Expected canonical:",
    cell67_after_bundle_fingerprint
)


print(
    "Derived:",
    cell75_active_derived_sha256
)


print(
    "Expected derived:",
    cell71_after_derived_bundle_sha256
)


print(
    "Fingerprint failures:",
    cell75_fingerprint_failures
)


print(
    "\n=== ACTIVE EXERCISE SLICE ==="
)


print(
    cell75_slice_counts
)


print(
    "Count failures:",
    cell75_slice_count_failures
)


print(
    "V2 identity failures:",
    cell75_v2_identity_failures
)


print(
    "\n=== TASK LIFECYCLE ==="
)


print(
    "construction_eligibility:",
    dict(
        cell75_task_construction_counts
    )
)


print(
    "target_evidence_status:",
    dict(
        cell75_task_target_status_counts
    )
)


print(
    "question_text_status:",
    dict(
        cell75_task_question_status_counts
    )
)


print(
    "expected_answer_status:",
    dict(
        cell75_task_expected_answer_status_counts
    )
)


print(
    "citation_target_status:",
    dict(
        cell75_task_citation_status_counts
    )
)


print(
    "Failures:",
    cell75_task_lifecycle_failures
)


print(
    "\n=== GROUNDABILITY SCHEMA DISCOVERY ==="
)


print(
    "Precheck field candidates:",
    cell75_groundability_precheck_field_candidates
)


print(
    "Resolved precheck field:",
    cell75_groundability_precheck_field
)


print(
    "construction_eligibility:",
    dict(
        cell75_groundability_construction_counts
    )
)


print(
    "precheck:",
    dict(
        cell75_groundability_precheck_counts
    )
)


print(
    "question_generation_status:",
    dict(
        cell75_groundability_question_counts
    )
)


print(
    "expected_answer_generation_status:",
    dict(
        cell75_groundability_answer_counts
    )
)


print(
    "Failures:",
    cell75_groundability_lifecycle_failures
)


print(
    "\n=== TASK DESIGN LIFECYCLE ==="
)


print(
    "task_construction_status:",
    dict(
        cell75_task_design_construction_counts
    )
)


print(
    "question_generation_status:",
    dict(
        cell75_task_design_question_counts
    )
)


print(
    "expected_answer_generation_status:",
    dict(
        cell75_task_design_answer_counts
    )
)


print(
    "Failures:",
    cell75_task_design_lifecycle_failures
)


print(
    "\n=== TARGET LIFECYCLE ==="
)


print(
    "evidence_target_method:",
    dict(
        cell75_target_method_counts
    )
)


print(
    "target_selection_basis:",
    dict(
        cell75_target_basis_counts
    )
)


print(
    "question_generation_status:",
    dict(
        cell75_target_question_counts
    )
)


print(
    "expected_answer_generation_status:",
    dict(
        cell75_target_answer_counts
    )
)


print(
    "Failures:",
    cell75_target_lifecycle_failures
)


print(
    "\n=== ACTIVE EXERCISE PILOT ==="
)


print(
    "Selection:",
    cell75_active_exercise_pilot_selection
)


print(
    "Question packet ID:",
    (
        cell75_active_exercise_question_packet.get(
            "pilot_case_id"
        )

        if (
            cell75_active_exercise_question_packet
            is not None
        )

        else None
    )
)


print(
    "Pilot failures:",
    cell75_pilot_failures
)


print(
    "\n=== LEGACY / HISTORICAL ==="
)


print(
    "Current-state Legacy leakage:",
    cell75_active_legacy_leakage
)


print(
    "Historical isolation failures:",
    cell75_historical_isolation_failures
)


print(
    "\n=== GENERATION STATE ==="
)


print(
    cell75_active_packet_generation_state
)


print(
    "Failures:",
    cell75_packet_generation_state_failures
)


print(
    "\n=== READ-ONLY STABILITY ==="
)


print(
    "Failure:",
    cell75_readonly_stability_failure
)


print(
    "\n=== CLOSURE ==="
)


print(
    "Closure failures:",
    cell75_closure_failures
)


print(
    "POST-ACTIVATION STATUS:",
    cell75_post_activation_status
)


print(
    "\n=== CODE CELL 75 GATES ==="
)


for gate_name, passed in (
    cell75_gates.items()
):

    print(
        gate_name,
        ":",
        "PASS"
        if passed
        else "FAIL"
    )


print(
    "\nFAILED GATES:",
    [
        gate_name

        for gate_name, passed
        in cell75_gates.items()

        if not passed
    ]
)


print(
    "\nCODE CELL 75 GATE:",
    "PASS"
    if cell_75_pass
    else "FAIL"
)


assert cell_75_pass, (
    "Exercise V2 activation closure failed. "
    "Do not begin question drafting. Repair this SAME "
    "Code Cell 75 using the printed schema-discovery, "
    "lifecycle, identity, fingerprint and isolation "
    "diagnostics."
)
```

```python
# Code Cell 76 — REPAIRED V2
#
# Runtime-grounded Question Drafting Contract Discovery
#
# READ ONLY.
# NO question generation.
# NO answer generation.
# NO citation adjudication.
# NO active mutation.
# NO artifact write.


# ============================================================
# 1) Helpers
# ============================================================

def cell76_sha(value):

    return hashlib.sha256(
        stable_json_key(value).encode("utf-8")
    ).hexdigest()


def cell76_clone(value):

    return cell52_safe_clone(value)


def cell76_field_counts(rows, field_name):

    return dict(
        Counter(
            row.get(
                field_name,
                "__MISSING__",
            )
            for row in rows
        )
    )


# ============================================================
# 2) Preconditions
# ============================================================

cell76_prerequisite_failures = []


if not cell_75_pass:

    cell76_prerequisite_failures.append(
        "CELL75_NOT_PASS"
    )


if (
    cell75_post_activation_status
    !=
    "EXERCISE_V2_ACTIVATION_CLOSED_"
    "READY_FOR_QUESTION_DRAFTING"
):

    cell76_prerequisite_failures.append(
        {
            "reason":
                "CELL75_STATUS_UNEXPECTED",

            "actual":
                cell75_post_activation_status,
        }
    )


# ============================================================
# 3) Active-state snapshot BEFORE discovery
# ============================================================

cell76_active_snapshot_before = {
    "canonical":
        cell76_sha(
            {
                "registries": {
                    "golden_case_records":
                        golden_case_records,

                    "auto_case_records":
                        auto_case_records,

                    "case_groundability_precheck":
                        case_groundability_precheck,

                    "case_task_design":
                        case_task_design,

                    "task_draft_plan_rows":
                        task_draft_plan_rows,

                    "target_evidence_records":
                        target_evidence_records,

                    "pilot_selection_rows":
                        pilot_selection_rows,
                },

                "indexes": {
                    "golden_case_records_by_id":
                        golden_case_records_by_id,

                    "task_draft_plans":
                        task_draft_plans,

                    "target_evidence_by_case_id":
                        target_evidence_by_case_id,

                    "target_evidence_by_task_draft_id":
                        target_evidence_by_task_draft_id,
                },
            }
        ),

    "derived":
        cell76_sha(
            {
                "candidate_hash_to_case_ids":
                    candidate_hash_to_case_ids,

                "cross_stratum_repeated_candidate_hashes":
                    cross_stratum_repeated_candidate_hashes,

                "question_drafting_pilot_packets":
                    question_drafting_pilot_packets,

                "question_drafting_pilot_by_id":
                    question_drafting_pilot_by_id,

                "target_id_counts":
                    target_id_counts,
            }
        ),
}


# ============================================================
# 4) Core runtime scope
# ============================================================

cell76_question_packet_count = (
    len(
        question_drafting_pilot_packets
    )
)


cell76_target_bearing_case_ids = {
    row[
        "golden_case_id"
    ]
    for row
    in target_evidence_records
}


cell76_packet_case_ids = {
    row[
        "golden_case_id"
    ]
    for row
    in question_drafting_pilot_packets
}


cell76_runtime_scope = {
    "question_packet_count":
        cell76_question_packet_count,

    "target_bearing_case_count":
        len(
            cell76_target_bearing_case_ids
        ),

    "packet_case_count":
        len(
            cell76_packet_case_ids
        ),

    "packet_cases_subset_of_targets":
        (
            cell76_packet_case_ids
            <=
            cell76_target_bearing_case_ids
        ),
}


cell76_question_workflow_scope = (
    "PILOT_ONLY_CURRENT_RUNTIME"
    if (
        cell76_question_packet_count
        ==
        23

        and

        len(
            cell76_target_bearing_case_ids
        )
        ==
        572

        and

        len(
            cell76_packet_case_ids
        )
        ==
        23
    )
    else
    "RUNTIME_SCOPE_REQUIRES_FURTHER_INSPECTION"
)


# ============================================================
# 5) Exact packet schema
# ============================================================

cell76_packet_schema_counter = Counter(
    tuple(
        sorted(
            packet.keys()
        )
    )
    for packet
    in question_drafting_pilot_packets
)


cell76_packet_schema_variants = [
    {
        "count":
            count,

        "fields":
            list(
                fields
            ),
    }

    for fields, count
    in cell76_packet_schema_counter.items()
]


cell76_packet_schema_uniform = (
    len(
        cell76_packet_schema_variants
    )
    ==
    1
)


cell76_packet_fields = (
    cell76_packet_schema_variants[
        0
    ][
        "fields"
    ]
    if (
        cell76_packet_schema_uniform
    )
    else []
)


# ============================================================
# 6) Status contract
# ============================================================

cell76_packet_status_fields = sorted(
    [
        field_name
        for field_name
        in cell76_packet_fields
        if (
            "status"
            in
            field_name.lower()
        )
    ]
)


cell76_packet_status_counts = {
    field_name:
        cell76_field_counts(
            question_drafting_pilot_packets,
            field_name,
        )

    for field_name
    in cell76_packet_status_fields
}


# ============================================================
# 7) Explicit pre-generation status inspection
# ============================================================

CELL76_EXPECTED_PREGEN_PACKET_FIELDS = [
    "question_draft_status",
    "question_groundability_status",
    "expected_answer_status",
    "citation_target_validation_status",
]


cell76_pregen_field_presence = {
    field_name:
        (
            field_name
            in
            cell76_packet_fields
        )

    for field_name
    in CELL76_EXPECTED_PREGEN_PACKET_FIELDS
}


cell76_pregen_status_failures = []


for field_name in (
    CELL76_EXPECTED_PREGEN_PACKET_FIELDS
):

    if (
        field_name
        not in
        cell76_packet_fields
    ):

        cell76_pregen_status_failures.append(
            {
                "field":
                    field_name,

                "reason":
                    "FIELD_MISSING",
            }
        )

        continue


    counts = (
        cell76_packet_status_counts[
            field_name
        ]
    )


    if (
        counts
        !=
        {
            "NOT_STARTED":
                23
        }
    ):

        cell76_pregen_status_failures.append(
            {
                "field":
                    field_name,

                "reason":
                    "STATUS_NOT_UNIFORMLY_NOT_STARTED",

                "actual":
                    counts,
            }
        )


# ============================================================
# 8) Guardrail inspection — TYPE AGNOSTIC
# ============================================================

cell76_guardrail_field_present_count = sum(
    1

    for packet
    in question_drafting_pilot_packets

    if (
        "guardrails"
        in
        packet
    )
)


cell76_guardrail_type_counts = dict(
    Counter(
        type(
            packet.get(
                "guardrails"
            )
        ).__name__

        for packet
        in question_drafting_pilot_packets
    )
)


cell76_guardrail_value_counts = dict(
    Counter(
        stable_json_key(
            packet.get(
                "guardrails"
            )
        )

        for packet
        in question_drafting_pilot_packets
    )
)


cell76_guardrail_examples = []


for packet in (
    question_drafting_pilot_packets
):

    serialized_guardrail = (
        stable_json_key(
            packet.get(
                "guardrails"
            )
        )
    )


    if (
        serialized_guardrail
        not in
        [
            row[
                "serialized"
            ]
            for row
            in cell76_guardrail_examples
        ]
    ):

        cell76_guardrail_examples.append(
            {
                "type":
                    type(
                        packet.get(
                            "guardrails"
                        )
                    ).__name__,

                "serialized":
                    serialized_guardrail,

                "value":
                    cell76_clone(
                        packet.get(
                            "guardrails"
                        )
                    ),
            }
        )


    if (
        len(
            cell76_guardrail_examples
        )
        >=
        5
    ):

        break


cell76_guardrail_contract = {
    "field_present_count":
        cell76_guardrail_field_present_count,

    "type_counts":
        cell76_guardrail_type_counts,

    "distinct_value_count":
        len(
            cell76_guardrail_value_counts
        ),

    "examples":
        cell76_guardrail_examples,
}


# ============================================================
# 9) Related lifecycle fields across runtime objects
# ============================================================

CELL76_RELATED_TERMS = (
    "question",
    "answer",
    "citation",
    "ground",
    "draft",
)


def cell76_related_fields(rows):

    return sorted(
        {
            key
            for row in rows
            for key in row.keys()
            if any(
                term
                in
                key.lower()
                for term
                in CELL76_RELATED_TERMS
            )
        }
    )


cell76_task_related_fields = (
    cell76_related_fields(
        task_draft_plan_rows
    )
)


cell76_groundability_related_fields = (
    cell76_related_fields(
        case_groundability_precheck
    )
)


cell76_design_related_fields = (
    cell76_related_fields(
        case_task_design
    )
)


cell76_target_related_fields = (
    cell76_related_fields(
        target_evidence_records
    )
)


# ============================================================
# 10) Source discovery
# ============================================================

CELL76_SOURCE_SEARCH_TERMS = [
    "question_drafting_pilot_packets",
    "question_drafting_pilot_by_id",
    "question_draft_status",
    "question_generation_status",
    "expected_answer",
    "citation_target",
    "question-pilot",
    "question_archetype",
]


cell76_source_occurrences = {
    term:
        []
    for term
    in CELL76_SOURCE_SEARCH_TERMS
}


for code_cell in (
    cell49_code_cells
):

    source_lines = (
        code_cell[
            "source"
        ].splitlines()
    )


    for term in (
        CELL76_SOURCE_SEARCH_TERMS
    ):

        for line_index, line in enumerate(
            source_lines
        ):

            if (
                term.lower()
                in
                line.lower()
            ):

                cell76_source_occurrences[
                    term
                ].append(
                    {
                        "cell_index":
                            code_cell[
                                "cell_index"
                            ],

                        "line_index":
                            line_index,

                        "line":
                            line.strip(),
                    }
                )


cell76_total_source_occurrence_count = sum(
    len(
        rows
    )
    for rows
    in cell76_source_occurrences.values()
)


# ============================================================
# 11) Function definitions related to generation
# ============================================================

cell76_function_candidates = []


for code_cell in (
    cell49_code_cells
):

    source_lines = (
        code_cell[
            "source"
        ].splitlines()
    )


    for line_index, line in enumerate(
        source_lines
    ):

        stripped = (
            line.strip()
        )


        if not (
            stripped.startswith(
                "def "
            )
        ):

            continue


        if any(
            term
            in
            stripped.lower()

            for term
            in (
                "question",
                "answer",
                "citation",
                "draft",
                "pilot",
            )
        ):

            cell76_function_candidates.append(
                {
                    "cell_index":
                        code_cell[
                            "cell_index"
                        ],

                    "line_index":
                        line_index,

                    "definition":
                        stripped,
                }
            )


# ============================================================
# 12) Detect REAL materialized question text
#
# Do NOT treat:
#   question_archetype
#   question_groundability_status
#   question_draft_status
# etc. as generated question text.
# ============================================================

CELL76_REAL_QUESTION_TEXT_FIELD_NAMES = {
    "question",
    "question_text",
    "question_draft",
    "question_draft_text",
    "generated_question",
    "generated_question_text",
}


cell76_candidate_question_text_fields = sorted(
    [
        field_name
        for field_name
        in cell76_packet_fields
        if (
            field_name.lower()
            in
            CELL76_REAL_QUESTION_TEXT_FIELD_NAMES
        )
    ]
)


cell76_materialized_question_text_rows = []


for packet in (
    question_drafting_pilot_packets
):

    for field_name in (
        cell76_candidate_question_text_fields
    ):

        value = (
            packet.get(
                field_name
            )
        )


        if (
            isinstance(
                value,
                str,
            )

            and

            value.strip()

            and

            value.strip()
            !=
            "NOT_STARTED"
        ):

            cell76_materialized_question_text_rows.append(
                {
                    "pilot_case_id":
                        packet.get(
                            "pilot_case_id"
                        ),

                    "field":
                        field_name,

                    "value":
                        value,
                }
            )


# ============================================================
# 13) Inspect Task-family / archetype distributions
# ============================================================

cell76_task_family_counts = dict(
    Counter(
        packet.get(
            "task_family"
        )
        for packet
        in question_drafting_pilot_packets
    )
)


cell76_question_archetype_counts = dict(
    Counter(
        packet.get(
            "question_archetype"
        )
        for packet
        in question_drafting_pilot_packets
    )
)


cell76_evidence_scope_counts = dict(
    Counter(
        packet.get(
            "evidence_scope"
        )
        for packet
        in question_drafting_pilot_packets
    )
)


# ============================================================
# 14) Discover packet ID contract evidence
# ============================================================

cell76_pilot_id_prefix_counts = dict(
    Counter(
        (
            packet.get(
                "pilot_case_id",
                ""
            ).split(
                ":",
                1,
            )[
                0
            ]
            if (
                ":"
                in
                packet.get(
                    "pilot_case_id",
                    ""
                )
            )
            else
            packet.get(
                "pilot_case_id"
            )
        )

        for packet
        in question_drafting_pilot_packets
    )
)


cell76_unique_pilot_ids = (
    len(
        {
            packet[
                "pilot_case_id"
            ]
            for packet
            in question_drafting_pilot_packets
        }
    )
)


# ============================================================
# 15) Discovery diagnostics / failures
# ============================================================

cell76_discovery_failures = []


if (
    cell76_question_packet_count
    !=
    23
):

    cell76_discovery_failures.append(
        {
            "reason":
                "QUESTION_PACKET_COUNT_NOT_23",

            "actual":
                cell76_question_packet_count,
        }
    )


if (
    len(
        cell76_target_bearing_case_ids
    )
    !=
    572
):

    cell76_discovery_failures.append(
        {
            "reason":
                "TARGET_BEARING_CASE_COUNT_NOT_572",

            "actual":
                len(
                    cell76_target_bearing_case_ids
                ),
        }
    )


if not (
    cell76_runtime_scope[
        "packet_cases_subset_of_targets"
    ]
):

    cell76_discovery_failures.append(
        "PACKET_CASES_NOT_SUBSET_OF_TARGET_CASES"
    )


if not (
    cell76_packet_schema_uniform
):

    cell76_discovery_failures.append(
        {
            "reason":
                "QUESTION_PACKET_SCHEMA_NOT_UNIFORM",

            "variant_count":
                len(
                    cell76_packet_schema_variants
                ),
        }
    )


if (
    cell76_pregen_status_failures
):

    cell76_discovery_failures.append(
        {
            "reason":
                "PREGENERATION_STATUS_FAILURE",

            "detail":
                cell76_pregen_status_failures,
        }
    )


if (
    cell76_guardrail_field_present_count
    !=
    23
):

    cell76_discovery_failures.append(
        {
            "reason":
                "GUARDRAILS_FIELD_NOT_PRESENT_ON_ALL_23_PACKETS",

            "actual":
                cell76_guardrail_field_present_count,
        }
    )


if (
    cell76_total_source_occurrence_count
    ==
    0
):

    cell76_discovery_failures.append(
        "NO_QUESTION_DRAFTING_SOURCE_EVIDENCE_FOUND"
    )


if (
    cell76_materialized_question_text_rows
):

    cell76_discovery_failures.append(
        {
            "reason":
                "QUESTION_TEXT_ALREADY_MATERIALIZED",

            "count":
                len(
                    cell76_materialized_question_text_rows
                ),
        }
    )


if (
    cell76_unique_pilot_ids
    !=
    23
):

    cell76_discovery_failures.append(
        {
            "reason":
                "QUESTION_PILOT_IDS_NOT_UNIQUE",

            "actual":
                cell76_unique_pilot_ids,
        }
    )


# ============================================================
# 16) Freeze discovery contract
# ============================================================

QUESTION_DRAFTING_CONTRACT_DISCOVERY_V1_DRAFT = {
    "version":
        "question_drafting_contract_discovery_v1_draft",

    "status":
        (
            "QUESTION_DRAFTING_CONTRACT_DISCOVERED_"
            "GENERATION_NOT_STARTED"

            if (
                not cell76_prerequisite_failures

                and

                not cell76_discovery_failures
            )

            else
            "QUESTION_DRAFTING_CONTRACT_DISCOVERY_FAILED"
        ),

    "runtime_scope":
        cell76_question_workflow_scope,

    "question_packet_count":
        cell76_question_packet_count,

    "target_bearing_case_count":
        len(
            cell76_target_bearing_case_ids
        ),

    "packet_schema":
        cell76_clone(
            cell76_packet_schema_variants
        ),

    "pregen_status_fields":
        list(
            CELL76_EXPECTED_PREGEN_PACKET_FIELDS
        ),

    "guardrail_contract":
        cell76_clone(
            cell76_guardrail_contract
        ),

    "task_family_counts":
        cell76_clone(
            cell76_task_family_counts
        ),

    "question_archetype_counts":
        cell76_clone(
            cell76_question_archetype_counts
        ),

    "evidence_scope_counts":
        cell76_clone(
            cell76_evidence_scope_counts
        ),

    "source_occurrence_count":
        cell76_total_source_occurrence_count,

    "function_candidate_count":
        len(
            cell76_function_candidates
        ),

    "real_question_text_fields":
        list(
            cell76_candidate_question_text_fields
        ),

    "materialized_question_text_count":
        len(
            cell76_materialized_question_text_rows
        ),

    "question_generation_status":
        "NOT_STARTED",

    "expected_answer_generation_status":
        "NOT_STARTED",

    "citation_adjudication_status":
        "NOT_STARTED",

    "artifact_write_status":
        "NONE",
}


# ============================================================
# 17) Final ACTIVE snapshot — must remain byte stable
# ============================================================

cell76_active_snapshot_after = {
    "canonical":
        cell76_sha(
            {
                "registries": {
                    "golden_case_records":
                        golden_case_records,

                    "auto_case_records":
                        auto_case_records,

                    "case_groundability_precheck":
                        case_groundability_precheck,

                    "case_task_design":
                        case_task_design,

                    "task_draft_plan_rows":
                        task_draft_plan_rows,

                    "target_evidence_records":
                        target_evidence_records,

                    "pilot_selection_rows":
                        pilot_selection_rows,
                },

                "indexes": {
                    "golden_case_records_by_id":
                        golden_case_records_by_id,

                    "task_draft_plans":
                        task_draft_plans,

                    "target_evidence_by_case_id":
                        target_evidence_by_case_id,

                    "target_evidence_by_task_draft_id":
                        target_evidence_by_task_draft_id,
                },
            }
        ),

    "derived":
        cell76_sha(
            {
                "candidate_hash_to_case_ids":
                    candidate_hash_to_case_ids,

                "cross_stratum_repeated_candidate_hashes":
                    cross_stratum_repeated_candidate_hashes,

                "question_drafting_pilot_packets":
                    question_drafting_pilot_packets,

                "question_drafting_pilot_by_id":
                    question_drafting_pilot_by_id,

                "target_id_counts":
                    target_id_counts,
            }
        ),
}


# ============================================================
# 18) Gates
# ============================================================

cell76_gates = {
    "exercise_v2_activation_closure_is_valid":
        (
            not cell76_prerequisite_failures
        ),

    "runtime_has_exactly_23_question_drafting_packets":
        (
            cell76_question_packet_count
            ==
            23
        ),

    "runtime_has_exactly_572_target_bearing_cases":
        (
            len(
                cell76_target_bearing_case_ids
            )
            ==
            572
        ),

    "question_drafting_runtime_scope_is_pilot_only":
        (
            cell76_question_workflow_scope
            ==
            "PILOT_ONLY_CURRENT_RUNTIME"
        ),

    "question_packet_schema_is_uniform":
        (
            cell76_packet_schema_uniform
        ),

    "all_four_pregeneration_status_fields_exist_and_are_not_started":
        (
            not cell76_pregen_status_failures
        ),

    "guardrails_field_is_present_on_all_23_packets":
        (
            cell76_guardrail_field_present_count
            ==
            23
        ),

    "question_drafting_source_evidence_exists_in_notebook":
        (
            cell76_total_source_occurrence_count
            >
            0
        ),

    "all_23_question_pilot_ids_are_unique":
        (
            cell76_unique_pilot_ids
            ==
            23
        ),

    "no_real_question_text_has_been_materialized":
        (
            len(
                cell76_materialized_question_text_rows
            )
            ==
            0
        ),

    "discovery_has_zero_failures":
        (
            not cell76_discovery_failures
        ),

    "question_drafting_contract_is_frozen_without_generation":
        (
            QUESTION_DRAFTING_CONTRACT_DISCOVERY_V1_DRAFT[
                "status"
            ]
            ==
            "QUESTION_DRAFTING_CONTRACT_DISCOVERED_"
            "GENERATION_NOT_STARTED"
        ),

    "active_state_remains_byte_stable":
        (
            cell76_active_snapshot_before
            ==
            cell76_active_snapshot_after
        ),

    "question_generation_not_started":
        (
            QUESTION_DRAFTING_CONTRACT_DISCOVERY_V1_DRAFT[
                "question_generation_status"
            ]
            ==
            "NOT_STARTED"
        ),

    "expected_answer_generation_not_started":
        (
            QUESTION_DRAFTING_CONTRACT_DISCOVERY_V1_DRAFT[
                "expected_answer_generation_status"
            ]
            ==
            "NOT_STARTED"
        ),

    "citation_adjudication_not_started":
        (
            QUESTION_DRAFTING_CONTRACT_DISCOVERY_V1_DRAFT[
                "citation_adjudication_status"
            ]
            ==
            "NOT_STARTED"
        ),

    "no_artifact_write":
        (
            QUESTION_DRAFTING_CONTRACT_DISCOVERY_V1_DRAFT[
                "artifact_write_status"
            ]
            ==
            "NONE"
        ),
}


cell_76_pass = all(
    cell76_gates.values()
)


# ============================================================
# 19) Output
# ============================================================

print(
    "=== QUESTION DRAFTING CONTRACT DISCOVERY — REPAIRED V2 ==="
)


print(
    "\n=== PREREQUISITES ==="
)


print(
    "Failures:",
    cell76_prerequisite_failures
)


print(
    "\n=== RUNTIME SCOPE ==="
)


print(
    cell76_runtime_scope
)


print(
    "Workflow scope:",
    cell76_question_workflow_scope
)


print(
    "\n=== PACKET SCHEMA ==="
)


print(
    "Uniform:",
    cell76_packet_schema_uniform
)


print(
    "Variants:",
    len(
        cell76_packet_schema_variants
    )
)


for variant in (
    cell76_packet_schema_variants
):

    print(
        "\nCount:",
        variant[
            "count"
        ]
    )


    print(
        "Fields:",
        variant[
            "fields"
        ]
    )


print(
    "\n=== PRE-GENERATION STATUS ==="
)


for field_name in (
    CELL76_EXPECTED_PREGEN_PACKET_FIELDS
):

    print(
        "-",
        field_name,
        ":",
        cell76_packet_status_counts.get(
            field_name
        )
    )


print(
    "Failures:",
    cell76_pregen_status_failures
)


print(
    "\n=== GUARDRAILS — TYPE AGNOSTIC ==="
)


print(
    "Present on packets:",
    cell76_guardrail_field_present_count,
    "/ 23"
)


print(
    "Type counts:",
    cell76_guardrail_type_counts
)


print(
    "Distinct values:",
    len(
        cell76_guardrail_value_counts
    )
)


print(
    "Examples:"
)


for example in (
    cell76_guardrail_examples
):

    print(
        "- type:",
        example[
            "type"
        ]
    )


    print(
        "  value:",
        example[
            "value"
        ]
    )


print(
    "\n=== TASK CONTRACT DISTRIBUTIONS ==="
)


print(
    "Task families:",
    cell76_task_family_counts
)


print(
    "Question archetypes:",
    cell76_question_archetype_counts
)


print(
    "Evidence scopes:",
    cell76_evidence_scope_counts
)


print(
    "\n=== RELATED LIFECYCLE FIELDS ==="
)


print(
    "Tasks:",
    cell76_task_related_fields
)


print(
    "Groundability:",
    cell76_groundability_related_fields
)


print(
    "Task Design:",
    cell76_design_related_fields
)


print(
    "Targets:",
    cell76_target_related_fields
)


print(
    "\n=== NOTEBOOK SOURCE EVIDENCE ==="
)


print(
    "Total occurrences:",
    cell76_total_source_occurrence_count
)


for term, rows in (
    cell76_source_occurrences.items()
):

    print(
        "-",
        term,
        ":",
        len(
            rows
        ),
        "cells=",
        sorted(
            {
                row[
                    "cell_index"
                ]
                for row
                in rows
            }
        ),
    )


print(
    "\nFunction candidates:",
    len(
        cell76_function_candidates
    )
)


for row in (
    cell76_function_candidates
):

    print(
        "- CELL",
        row[
            "cell_index"
        ],
        "LINE",
        row[
            "line_index"
        ],
        ":",
        row[
            "definition"
        ]
    )


print(
    "\n=== REAL QUESTION TEXT DETECTION ==="
)


print(
    "Candidate text fields:",
    cell76_candidate_question_text_fields
)


print(
    "Materialized count:",
    len(
        cell76_materialized_question_text_rows
    )
)


print(
    "Rows:",
    cell76_materialized_question_text_rows[
        :5
    ]
)


print(
    "\n=== PILOT ID CONTRACT ==="
)


print(
    "Prefix counts:",
    cell76_pilot_id_prefix_counts
)


print(
    "Unique IDs:",
    cell76_unique_pilot_ids
)


print(
    "\n=== DISCOVERY FAILURES ==="
)


print(
    cell76_discovery_failures
)


print(
    "\nDISCOVERY STATUS:",
    QUESTION_DRAFTING_CONTRACT_DISCOVERY_V1_DRAFT[
        "status"
    ]
)


print(
    "\n=== CODE CELL 76 GATES ==="
)


for gate_name, passed in (
    cell76_gates.items()
):

    print(
        gate_name,
        ":",
        "PASS"
        if passed
        else "FAIL"
    )


print(
    "\nFAILED GATES:",
    [
        gate_name
        for gate_name, passed
        in cell76_gates.items()
        if not passed
    ]
)


print(
    "\nCODE CELL 76 GATE:",
    "PASS"
    if cell_76_pass
    else "FAIL"
)


assert cell_76_pass, (
    "Question Drafting contract discovery is still not "
    "fully resolved. Do not generate questions. Use the "
    "printed runtime scope, exact packet schema, status, "
    "guardrail type, source evidence and real-question-text "
    "diagnostics to repair this SAME Code Cell 76."
)
```

<h3>Pilot Question-Drafting Contract Freeze</h3>

<p>
Cell 77 ثابت کرد که Runtime فعلی برای 23 Pilot Packet آماده است، اما هیچ Question Builder اجرایی در Notebook وجود ندارد.
بنابراین قبل از تولید حتی یک سؤال، باید Contract رسمی Drafting تعریف و Validate شود.
</p>

<p>
این Contract فقط برای 23 Pilot فعلی است و شامل موارد زیر می‌شود:
</p>

<ul>
<li>هر Drafting Unit دقیقاً از یک Question Pilot Packet معتبر شروع می‌شود؛</li>
<li>Question فقط بر اساس <code>target_text_parts</code> و Context/Authority موجود در همان Packet ساخته می‌شود؛</li>
<li>External Knowledge، Speaker Inference، Unsupported Attribution، Locator Fabrication، Answer Leakage و Internal ID Leakage ممنوع هستند؛</li>
<li>Question باید با <code>task_family</code>، <code>question_archetype</code> و <code>required_dimensions</code> همان Pilot سازگار باشد؛</li>
<li>Expected Answer و Citation Adjudication در مرحله‌ی Draft Question تولید نمی‌شوند؛</li>
<li>هیچ Status فعالی در این Cell تغییر نمی‌کند؛</li>
<li>گسترش از 23 Pilot به 572 Case تا قبل از Pilot Validation مجاز نیست.</li>
</ul>

<p>
اگر این Cell PASS شود، Contract تولید سؤال Pilot به‌صورت رسمی Freeze شده و مرحله‌ی بعد می‌تواند طراحی Drafting Template / Rubric واقعی را آغاز کند.
</p>

</div>

```python
# Code Cell 78

# ============================================================
# Pilot Question-Drafting Contract Freeze
#
# Cell77 established:
#
#   - 23 Pilot packets
#   - no executable Question builder
#   - all generation statuses NOT_STARTED
#   - next step =
#       DEFINE_AND_VALIDATE_PILOT_QUESTION_DRAFTING_CONTRACT
#
# This Cell defines that contract WITHOUT generating text.
#
# READ-ONLY against ACTIVE state.
#
# NO question text.
# NO expected answer.
# NO citation adjudication.
# NO active mutation.
# NO artifact write.
# ============================================================


# ============================================================
# 1) Helpers
# ============================================================

def cell78_sha(value):

    return hashlib.sha256(
        stable_json_key(
            value
        ).encode(
            "utf-8"
        )
    ).hexdigest()


def cell78_clone(value):

    return cell52_safe_clone(
        value
    )


# ============================================================
# 2) Preconditions
# ============================================================

cell78_prerequisite_failures = []


if not (
    cell_77_pass
):

    cell78_prerequisite_failures.append(
        "CELL77_NOT_PASS"
    )


if (
    cell77_next_step_category
    !=
    "DEFINE_AND_VALIDATE_PILOT_QUESTION_DRAFTING_CONTRACT"
):

    cell78_prerequisite_failures.append(
        {
            "reason":
                "CELL77_NEXT_STEP_UNEXPECTED",

            "actual":
                cell77_next_step_category,
        }
    )


if (
    QUESTION_DRAFTING_WORKFLOW_RESOLUTION_V1_DRAFT[
        "status"
    ]
    !=
    "PILOT_QUESTION_DRAFTING_WORKFLOW_RESOLVED_"
    "GENERATION_STILL_BLOCKED"
):

    cell78_prerequisite_failures.append(
        {
            "reason":
                "CELL77_WORKFLOW_STATUS_UNEXPECTED",

            "actual":
                QUESTION_DRAFTING_WORKFLOW_RESOLUTION_V1_DRAFT[
                    "status"
                ],
        }
    )


# ============================================================
# 3) Snapshot ACTIVE state
# ============================================================

cell78_active_snapshot_before = {
    "canonical":
        cell78_sha(
            {
                "golden_case_records":
                    golden_case_records,

                "case_groundability_precheck":
                    case_groundability_precheck,

                "case_task_design":
                    case_task_design,

                "task_draft_plan_rows":
                    task_draft_plan_rows,

                "target_evidence_records":
                    target_evidence_records,

                "pilot_selection_rows":
                    pilot_selection_rows,

                "golden_case_records_by_id":
                    golden_case_records_by_id,

                "task_draft_plans":
                    task_draft_plans,

                "target_evidence_by_case_id":
                    target_evidence_by_case_id,

                "target_evidence_by_task_draft_id":
                    target_evidence_by_task_draft_id,
            }
        ),

    "pilot_runtime":
        cell78_sha(
            {
                "question_drafting_pilot_packets":
                    question_drafting_pilot_packets,

                "question_drafting_pilot_by_id":
                    question_drafting_pilot_by_id,
            }
        ),
}


# ============================================================
# 4) Required input fields
#
# These fields exist uniformly on all 23 packets according
# to Cell76.
# ============================================================

CELL78_REQUIRED_PACKET_INPUT_FIELDS = [
    "pilot_case_id",
    "golden_case_id",
    "task_draft_id",
    "target_evidence_id",
    "source_type",
    "stratum",
    "task_family",
    "question_archetype",
    "required_dimensions",
    "authority_expectation",
    "context_expectation",
    "evidence_scope",
    "target_selection_basis",
    "target_selection_method",
    "target_text_parts",
    "target_total_tokens",
    "candidate_identity",
    "guardrails",
]


cell78_missing_required_field_failures = []


for packet in (
    question_drafting_pilot_packets
):

    missing_fields = [
        field_name

        for field_name
        in CELL78_REQUIRED_PACKET_INPUT_FIELDS

        if (
            field_name
            not in
            packet
        )
    ]


    if missing_fields:

        cell78_missing_required_field_failures.append(
            {
                "pilot_case_id":
                    packet.get(
                        "pilot_case_id"
                    ),

                "missing_fields":
                    missing_fields,
            }
        )


# ============================================================
# 5) Required pre-generation statuses
# ============================================================

CELL78_REQUIRED_PREGEN_STATUSES = {
    "question_draft_status":
        "NOT_STARTED",

    "question_groundability_status":
        "NOT_STARTED",

    "expected_answer_status":
        "NOT_STARTED",

    "citation_target_validation_status":
        "NOT_STARTED",
}


cell78_pregen_status_failures = []


for packet in (
    question_drafting_pilot_packets
):

    for field_name, expected_value in (
        CELL78_REQUIRED_PREGEN_STATUSES.items()
    ):

        actual_value = (
            packet.get(
                field_name
            )
        )


        if (
            actual_value
            !=
            expected_value
        ):

            cell78_pregen_status_failures.append(
                {
                    "pilot_case_id":
                        packet.get(
                            "pilot_case_id"
                        ),

                    "field":
                        field_name,

                    "expected":
                        expected_value,

                    "actual":
                        actual_value,
                }
            )


# ============================================================
# 6) Freeze exact runtime guardrail contract
#
# Cell76 proved a single common Guardrail dictionary across
# all 23 packets.
# ============================================================

cell78_guardrail_variants = {}


for packet in (
    question_drafting_pilot_packets
):

    guardrails = (
        packet.get(
            "guardrails"
        )
    )


    guardrail_sha = (
        cell78_sha(
            guardrails
        )
    )


    if (
        guardrail_sha
        not in
        cell78_guardrail_variants
    ):

        cell78_guardrail_variants[
            guardrail_sha
        ] = {
            "value":
                cell78_clone(
                    guardrails
                ),

            "count":
                0,
        }


    cell78_guardrail_variants[
        guardrail_sha
    ][
        "count"
    ] += 1


cell78_guardrail_failures = []


if (
    len(
        cell78_guardrail_variants
    )
    !=
    1
):

    cell78_guardrail_failures.append(
        {
            "reason":
                "GUARDRAIL_CONTRACT_NOT_UNIFORM",

            "variant_count":
                len(
                    cell78_guardrail_variants
                ),
        }
    )


cell78_frozen_guardrails = (
    cell78_clone(
        next(
            iter(
                cell78_guardrail_variants.values()
            )
        )[
            "value"
        ]
    )

    if (
        len(
            cell78_guardrail_variants
        )
        ==
        1
    )

    else None
)


CELL78_REQUIRED_GUARDRAILS = {
    "external_knowledge":
        "FORBIDDEN",

    "unknown_speaker_inference":
        "FORBIDDEN",

    "unsupported_attribution":
        "FORBIDDEN",

    "secondary_to_primary_promotion":
        "FORBIDDEN",

    "locator_fabrication":
        "FORBIDDEN",

    "answer_leakage_in_question":
        "FORBIDDEN",

    "internal_ids_in_question":
        "FORBIDDEN",
}


if (
    cell78_frozen_guardrails
    !=
    CELL78_REQUIRED_GUARDRAILS
):

    cell78_guardrail_failures.append(
        {
            "reason":
                "FROZEN_GUARDRAILS_DO_NOT_MATCH_RUNTIME",

            "expected":
                CELL78_REQUIRED_GUARDRAILS,

            "actual":
                cell78_frozen_guardrails,
        }
    )


# ============================================================
# 7) Evidence availability
#
# A drafting unit cannot be admitted unless it has actual
# target text.
# ============================================================

cell78_evidence_failures = []


for packet in (
    question_drafting_pilot_packets
):

    target_parts = (
        packet.get(
            "target_text_parts"
        )
    )


    if not (
        isinstance(
            target_parts,
            list,
        )

        and

        len(
            target_parts
        )
        >
        0
    ):

        cell78_evidence_failures.append(
            {
                "pilot_case_id":
                    packet.get(
                        "pilot_case_id"
                    ),

                "reason":
                    "TARGET_TEXT_PARTS_EMPTY_OR_INVALID",
            }
        )

        continue


    nonempty_text_count = sum(
        1

        for part
        in target_parts

        if (
            isinstance(
                part,
                dict,
            )

            and

            isinstance(
                part.get(
                    "text"
                ),
                str,
            )

            and

            part.get(
                "text"
            ).strip()
        )
    )


    if (
        nonempty_text_count
        !=
        len(
            target_parts
        )
    ):

        cell78_evidence_failures.append(
            {
                "pilot_case_id":
                    packet.get(
                        "pilot_case_id"
                    ),

                "reason":
                    "ONE_OR_MORE_TARGET_PARTS_HAVE_NO_TEXT",

                "part_count":
                    len(
                        target_parts
                    ),

                "nonempty_text_count":
                    nonempty_text_count,
            }
        )


# ============================================================
# 8) Task / archetype / dimension presence
# ============================================================

cell78_semantic_contract_failures = []


for packet in (
    question_drafting_pilot_packets
):

    for field_name in (
        [
            "task_family",
            "question_archetype",
            "required_dimensions",
            "authority_expectation",
            "context_expectation",
            "evidence_scope",
        ]
    ):

        value = (
            packet.get(
                field_name
            )
        )


        if (
            value is None

            or

            value == ""

            or

            value == []
        ):

            cell78_semantic_contract_failures.append(
                {
                    "pilot_case_id":
                        packet.get(
                            "pilot_case_id"
                        ),

                    "field":
                        field_name,

                    "reason":
                        "SEMANTIC_CONTRACT_FIELD_EMPTY",
                }
            )


# ============================================================
# 9) Drafting-unit identity contract
#
# Contract:
#   one CURRENT drafting unit corresponds to one current
#   Pilot Packet.
#
# This does NOT yet imply one final accepted question.
# ============================================================

cell78_drafting_unit_ids = [
    packet[
        "pilot_case_id"
    ]

    for packet
    in question_drafting_pilot_packets
]


cell78_identity_failures = []


if (
    len(
        cell78_drafting_unit_ids
    )
    !=
    23
):

    cell78_identity_failures.append(
        {
            "reason":
                "DRAFTING_UNIT_COUNT_NOT_23",

            "actual":
                len(
                    cell78_drafting_unit_ids
                ),
        }
    )


if (
    len(
        set(
            cell78_drafting_unit_ids
        )
    )
    !=
    23
):

    cell78_identity_failures.append(
        "DRAFTING_UNIT_IDS_NOT_UNIQUE"
    )


# ============================================================
# 10) Explicit stage separation
#
# Question drafting does NOT simultaneously create:
#
#   expected answer
#   citation adjudication
#   full-population generation
# ============================================================

CELL78_STAGE_SEPARATION_POLICY = {
    "question_drafting_scope":
        "23_PILOT_PACKETS_ONLY",

    "expected_answer_generation":
        "BLOCKED_UNTIL_QUESTION_DRAFT_VALIDATION",

    "citation_target_adjudication":
        "BLOCKED_UNTIL_QUESTION_DRAFT_VALIDATION",

    "full_572_case_generation":
        "BLOCKED_UNTIL_PILOT_WORKFLOW_VALIDATED",

    "external_knowledge":
        "FORBIDDEN",

    "active_state_mutation_in_contract_freeze":
        "FORBIDDEN",
}


# ============================================================
# 11) Source-grounded Drafting Contract
#
# IMPORTANT:
#
# This contract defines WHAT a future question draft must
# obey. It does NOT generate the question itself.
# ============================================================

PILOT_QUESTION_DRAFTING_CONTRACT_V1_DRAFT = {
    "version":
        "pilot_question_drafting_contract_v1_draft",

    "scope":
        "23_CURRENT_QUESTION_DRAFTING_PILOT_PACKETS",

    "drafting_unit":
        "ONE_CURRENT_QUESTION_PILOT_PACKET",

    "drafting_unit_count":
        23,

    "required_packet_input_fields":
        list(
            CELL78_REQUIRED_PACKET_INPUT_FIELDS
        ),

    "required_pre_generation_statuses":
        cell78_clone(
            CELL78_REQUIRED_PREGEN_STATUSES
        ),

    "evidence_policy": {
        "allowed_evidence":
            "TARGET_TEXT_PARTS_FROM_CURRENT_PILOT_PACKET",

        "external_knowledge":
            "FORBIDDEN",

        "evidence_scope_must_follow_packet":
            True,

        "authority_expectation_must_follow_packet":
            True,

        "context_expectation_must_follow_packet":
            True,
    },

    "question_semantic_policy": {
        "must_follow_task_family":
            True,

        "must_follow_question_archetype":
            True,

        "must_support_required_dimensions":
            True,

        "must_be_groundable_in_target_text_parts":
            True,

        "must_not_contain_answer":
            True,

        "must_not_expose_internal_ids":
            True,

        "must_not_fabricate_locator":
            True,

        "must_not_infer_unknown_speaker":
            True,

        "must_not_upgrade_secondary_authority":
            True,

        "must_not_make_unsupported_attribution":
            True,
    },

    "frozen_guardrails":
        cell78_clone(
            cell78_frozen_guardrails
        ),

    "stage_separation":
        cell78_clone(
            CELL78_STAGE_SEPARATION_POLICY
        ),

    "question_output_status_before_drafting":
        "NOT_STARTED",

    "expected_answer_status_before_drafting":
        "NOT_STARTED",

    "citation_validation_status_before_drafting":
        "NOT_STARTED",

    "authoring_mode":
        "NOT_YET_SELECTED",

    "question_text_schema":
        "NOT_YET_DEFINED",

    "question_id_schema":
        "NOT_YET_DEFINED",

    "acceptance_rubric":
        "NOT_YET_DEFINED",

    "generation_execution":
        "NOT_STARTED",

    "artifact_write":
        "NONE",
}


# ============================================================
# 12) Contract validation failures
# ============================================================

cell78_contract_validation_failures = []


if (
    PILOT_QUESTION_DRAFTING_CONTRACT_V1_DRAFT[
        "drafting_unit_count"
    ]
    !=
    23
):

    cell78_contract_validation_failures.append(
        "CONTRACT_DRAFTING_UNIT_COUNT_NOT_23"
    )


if (
    PILOT_QUESTION_DRAFTING_CONTRACT_V1_DRAFT[
        "authoring_mode"
    ]
    !=
    "NOT_YET_SELECTED"
):

    cell78_contract_validation_failures.append(
        "AUTHORING_MODE_WAS_GUESSED"
    )


if (
    PILOT_QUESTION_DRAFTING_CONTRACT_V1_DRAFT[
        "question_text_schema"
    ]
    !=
    "NOT_YET_DEFINED"
):

    cell78_contract_validation_failures.append(
        "QUESTION_TEXT_SCHEMA_WAS_PREMATURELY_DEFINED"
    )


if (
    PILOT_QUESTION_DRAFTING_CONTRACT_V1_DRAFT[
        "question_id_schema"
    ]
    !=
    "NOT_YET_DEFINED"
):

    cell78_contract_validation_failures.append(
        "QUESTION_ID_SCHEMA_WAS_PREMATURELY_DEFINED"
    )


if (
    PILOT_QUESTION_DRAFTING_CONTRACT_V1_DRAFT[
        "acceptance_rubric"
    ]
    !=
    "NOT_YET_DEFINED"
):

    cell78_contract_validation_failures.append(
        "ACCEPTANCE_RUBRIC_WAS_PREMATURELY_DEFINED"
    )


# ============================================================
# 13) Freeze readiness state
#
# We have now frozen the boundary contract, but generation
# remains blocked until:
#
#   authoring mode
#   question schema
#   question ID schema
#   acceptance rubric
#
# are explicitly designed and validated.
# ============================================================

cell78_contract_ready = (
    len(
        cell78_prerequisite_failures
    )
    ==
    0

    and

    len(
        cell78_missing_required_field_failures
    )
    ==
    0

    and

    len(
        cell78_pregen_status_failures
    )
    ==
    0

    and

    len(
        cell78_guardrail_failures
    )
    ==
    0

    and

    len(
        cell78_evidence_failures
    )
    ==
    0

    and

    len(
        cell78_semantic_contract_failures
    )
    ==
    0

    and

    len(
        cell78_identity_failures
    )
    ==
    0

    and

    len(
        cell78_contract_validation_failures
    )
    ==
    0
)


cell78_contract_status = (
    "PILOT_QUESTION_DRAFTING_BOUNDARY_CONTRACT_FROZEN_"
    "GENERATION_BLOCKED_PENDING_AUTHORING_SCHEMA_AND_RUBRIC"

    if (
        cell78_contract_ready
    )

    else
    "PILOT_QUESTION_DRAFTING_CONTRACT_FREEZE_FAILED"
)


# ============================================================
# 14) Active-state snapshot AFTER
# ============================================================

cell78_active_snapshot_after = {
    "canonical":
        cell78_sha(
            {
                "golden_case_records":
                    golden_case_records,

                "case_groundability_precheck":
                    case_groundability_precheck,

                "case_task_design":
                    case_task_design,

                "task_draft_plan_rows":
                    task_draft_plan_rows,

                "target_evidence_records":
                    target_evidence_records,

                "pilot_selection_rows":
                    pilot_selection_rows,

                "golden_case_records_by_id":
                    golden_case_records_by_id,

                "task_draft_plans":
                    task_draft_plans,

                "target_evidence_by_case_id":
                    target_evidence_by_case_id,

                "target_evidence_by_task_draft_id":
                    target_evidence_by_task_draft_id,
            }
        ),

    "pilot_runtime":
        cell78_sha(
            {
                "question_drafting_pilot_packets":
                    question_drafting_pilot_packets,

                "question_drafting_pilot_by_id":
                    question_drafting_pilot_by_id,
            }
        ),
}


# ============================================================
# 15) Gates
# ============================================================

cell78_gates = {
    "cell77_workflow_resolution_is_pass":
        (
            len(
                cell78_prerequisite_failures
            )
            ==
            0
        ),

    "all_23_pilot_packets_have_required_contract_inputs":
        (
            len(
                cell78_missing_required_field_failures
            )
            ==
            0
        ),

    "all_23_pilot_packets_are_in_exact_pregeneration_state":
        (
            len(
                cell78_pregen_status_failures
            )
            ==
            0
        ),

    "runtime_guardrail_contract_is_uniform_and_frozen":
        (
            len(
                cell78_guardrail_failures
            )
            ==
            0

            and

            cell78_frozen_guardrails
            ==
            CELL78_REQUIRED_GUARDRAILS
        ),

    "all_23_pilot_packets_have_nonempty_target_evidence":
        (
            len(
                cell78_evidence_failures
            )
            ==
            0
        ),

    "all_23_pilot_packets_have_complete_semantic_task_contracts":
        (
            len(
                cell78_semantic_contract_failures
            )
            ==
            0
        ),

    "exactly_23_unique_drafting_units_are_defined":
        (
            len(
                cell78_identity_failures
            )
            ==
            0
        ),

    "question_expected_answer_and_citation_stages_are_explicitly_separated":
        (
            CELL78_STAGE_SEPARATION_POLICY[
                "expected_answer_generation"
            ]
            ==
            "BLOCKED_UNTIL_QUESTION_DRAFT_VALIDATION"

            and

            CELL78_STAGE_SEPARATION_POLICY[
                "citation_target_adjudication"
            ]
            ==
            "BLOCKED_UNTIL_QUESTION_DRAFT_VALIDATION"
        ),

    "full_572_case_generation_remains_blocked_until_pilot_validation":
        (
            CELL78_STAGE_SEPARATION_POLICY[
                "full_572_case_generation"
            ]
            ==
            "BLOCKED_UNTIL_PILOT_WORKFLOW_VALIDATED"
        ),

    "authoring_mode_is_not_guessed":
        (
            PILOT_QUESTION_DRAFTING_CONTRACT_V1_DRAFT[
                "authoring_mode"
            ]
            ==
            "NOT_YET_SELECTED"
        ),

    "question_schema_and_acceptance_rubric_are_not_prematurely_invented":
        (
            PILOT_QUESTION_DRAFTING_CONTRACT_V1_DRAFT[
                "question_text_schema"
            ]
            ==
            "NOT_YET_DEFINED"

            and

            PILOT_QUESTION_DRAFTING_CONTRACT_V1_DRAFT[
                "question_id_schema"
            ]
            ==
            "NOT_YET_DEFINED"

            and

            PILOT_QUESTION_DRAFTING_CONTRACT_V1_DRAFT[
                "acceptance_rubric"
            ]
            ==
            "NOT_YET_DEFINED"
        ),

    "boundary_contract_is_frozen":
        (
            cell78_contract_status
            ==
            "PILOT_QUESTION_DRAFTING_BOUNDARY_CONTRACT_FROZEN_"
            "GENERATION_BLOCKED_PENDING_AUTHORING_SCHEMA_AND_RUBRIC"
        ),

    "active_state_remains_byte_stable":
        (
            cell78_active_snapshot_before
            ==
            cell78_active_snapshot_after
        ),

    "no_question_generation":
        (
            PILOT_QUESTION_DRAFTING_CONTRACT_V1_DRAFT[
                "generation_execution"
            ]
            ==
            "NOT_STARTED"
        ),

    "no_artifact_write":
        (
            PILOT_QUESTION_DRAFTING_CONTRACT_V1_DRAFT[
                "artifact_write"
            ]
            ==
            "NONE"
        ),
}


cell_78_pass = all(
    cell78_gates.values()
)


# ============================================================
# 16) Output
# ============================================================

print(
    "=== PILOT QUESTION-DRAFTING CONTRACT FREEZE ==="
)


print(
    "\n=== PREREQUISITES ==="
)


print(
    "Failures:",
    cell78_prerequisite_failures
)


print(
    "\n=== DRAFTING UNITS ==="
)


print(
    "Count:",
    len(
        cell78_drafting_unit_ids
    )
)


print(
    "Unique:",
    len(
        set(
            cell78_drafting_unit_ids
        )
    )
)


print(
    "\n=== REQUIRED INPUT CONTRACT ==="
)


print(
    CELL78_REQUIRED_PACKET_INPUT_FIELDS
)


print(
    "Missing-field failures:",
    cell78_missing_required_field_failures
)


print(
    "\n=== PRE-GENERATION STATUS CONTRACT ==="
)


print(
    CELL78_REQUIRED_PREGEN_STATUSES
)


print(
    "Failures:",
    cell78_pregen_status_failures
)


print(
    "\n=== FROZEN GUARDRAILS ==="
)


print(
    cell78_frozen_guardrails
)


print(
    "Guardrail failures:",
    cell78_guardrail_failures
)


print(
    "\n=== EVIDENCE VALIDATION ==="
)


print(
    "Failures:",
    cell78_evidence_failures
)


print(
    "\n=== SEMANTIC TASK CONTRACT ==="
)


print(
    "Failures:",
    cell78_semantic_contract_failures
)


print(
    "\n=== STAGE SEPARATION ==="
)


print(
    CELL78_STAGE_SEPARATION_POLICY
)


print(
    "\n=== UNRESOLVED DESIGN ITEMS ==="
)


print(
    "Authoring mode:",
    PILOT_QUESTION_DRAFTING_CONTRACT_V1_DRAFT[
        "authoring_mode"
    ]
)


print(
    "Question text schema:",
    PILOT_QUESTION_DRAFTING_CONTRACT_V1_DRAFT[
        "question_text_schema"
    ]
)


print(
    "Question ID schema:",
    PILOT_QUESTION_DRAFTING_CONTRACT_V1_DRAFT[
        "question_id_schema"
    ]
)


print(
    "Acceptance rubric:",
    PILOT_QUESTION_DRAFTING_CONTRACT_V1_DRAFT[
        "acceptance_rubric"
    ]
)


print(
    "\nContract validation failures:",
    cell78_contract_validation_failures
)


print(
    "\nCONTRACT STATUS:",
    cell78_contract_status
)


print(
    "\n=== CODE CELL 78 GATES ==="
)


for gate_name, passed in (
    cell78_gates.items()
):

    print(
        gate_name,
        ":",
        "PASS"
        if passed
        else "FAIL"
    )


print(
    "\nFAILED GATES:",
    [
        gate_name

        for gate_name, passed
        in cell78_gates.items()

        if not passed
    ]
)


print(
    "\nCODE CELL 78 GATE:",
    "PASS"
    if cell_78_pass
    else "FAIL"
)


assert cell_78_pass, (
    "Pilot Question-Drafting boundary contract could not be "
    "frozen safely. Do not generate questions. Repair this "
    "SAME Code Cell 78 using the printed input, status, "
    "guardrail, evidence and stage-separation diagnostics."
)
```

```python
# Code Cell 79

# ============================================================
# Final Design Evidence Discovery
#
# Goal:
#   resolve actual Notebook/runtime evidence for:
#
#   1) Authoring Mode
#   2) Question Text Schema
#   3) Question ID Schema
#   4) Acceptance Rubric
#
# before defining any of them.
#
# READ ONLY.
# NO question generation.
# NO answer generation.
# NO citation adjudication.
# NO active mutation.
# NO artifact write.
# ============================================================


# ============================================================
# 1) Helpers
# ============================================================

def cell79_sha(value):

    return hashlib.sha256(
        stable_json_key(
            value
        ).encode(
            "utf-8"
        )
    ).hexdigest()


def cell79_clone(value):

    return cell52_safe_clone(
        value
    )


def cell79_source_contexts(
    search_terms,
    context_before=12,
    context_after=18,
):

    results = []


    for code_cell in (
        cell49_code_cells
    ):

        source_lines = (
            code_cell[
                "source"
            ].splitlines()
        )


        for line_index, line in enumerate(
            source_lines
        ):

            lowered = (
                line.lower()
            )


            matched_terms = [
                term

                for term
                in search_terms

                if (
                    term.lower()
                    in
                    lowered
                )
            ]


            if not matched_terms:

                continue


            start = max(
                0,
                line_index - context_before,
            )


            end = min(
                len(
                    source_lines
                ),
                line_index + context_after + 1,
            )


            results.append(
                {
                    "cell_index":
                        code_cell[
                            "cell_index"
                        ],

                    "line_index":
                        line_index,

                    "matched_terms":
                        matched_terms,

                    "line":
                        line.strip(),

                    "snippet":
                        "\n".join(
                            source_lines[
                                start:end
                            ]
                        ),
                }
            )


    return results


# ============================================================
# 2) Preconditions
# ============================================================

cell79_prerequisite_failures = []


if not (
    cell_78_pass
):

    cell79_prerequisite_failures.append(
        "CELL78_NOT_PASS"
    )


if (
    cell78_contract_status
    !=
    "PILOT_QUESTION_DRAFTING_BOUNDARY_CONTRACT_FROZEN_"
    "GENERATION_BLOCKED_PENDING_AUTHORING_SCHEMA_AND_RUBRIC"
):

    cell79_prerequisite_failures.append(
        {
            "reason":
                "CELL78_CONTRACT_STATUS_UNEXPECTED",

            "actual":
                cell78_contract_status,
        }
    )


# ============================================================
# 3) Active-state snapshot BEFORE
# ============================================================

cell79_active_snapshot_before = {
    "canonical":
        cell79_sha(
            {
                "golden_case_records":
                    golden_case_records,

                "case_groundability_precheck":
                    case_groundability_precheck,

                "case_task_design":
                    case_task_design,

                "task_draft_plan_rows":
                    task_draft_plan_rows,

                "target_evidence_records":
                    target_evidence_records,

                "pilot_selection_rows":
                    pilot_selection_rows,

                "golden_case_records_by_id":
                    golden_case_records_by_id,

                "task_draft_plans":
                    task_draft_plans,

                "target_evidence_by_case_id":
                    target_evidence_by_case_id,

                "target_evidence_by_task_draft_id":
                    target_evidence_by_task_draft_id,
            }
        ),

    "pilot_runtime":
        cell79_sha(
            {
                "question_drafting_pilot_packets":
                    question_drafting_pilot_packets,

                "question_drafting_pilot_by_id":
                    question_drafting_pilot_by_id,
            }
        ),
}


# ============================================================
# 4) AUTHORING MODE evidence
# ============================================================

CELL79_AUTHORING_MODE_SEARCH_TERMS = [
    "human",
    "manual",
    "model",
    "llm",
    "authoring",
    "drafting",
    "generate",
    "generation",
    "review",
    "adjudicat",
]


cell79_authoring_mode_source_contexts = (
    cell79_source_contexts(
        CELL79_AUTHORING_MODE_SEARCH_TERMS,
        context_before=15,
        context_after=25,
    )
)


cell79_authoring_mode_exact_signal_counts = {
    term:
        sum(
            1

            for row
            in cell79_authoring_mode_source_contexts

            if (
                term
                in
                [
                    matched.lower()

                    for matched
                    in row[
                        "matched_terms"
                    ]
                ]
            )
        )

    for term
    in CELL79_AUTHORING_MODE_SEARCH_TERMS
}


# ============================================================
# 5) QUESTION TEXT SCHEMA evidence
# ============================================================

CELL79_QUESTION_SCHEMA_SEARCH_TERMS = [
    "question_text",
    "question text",
    "question_draft",
    "question draft",
    "question_schema",
    "question schema",
    "question_record",
    "question record",
    "question_type",
    "question type",
]


cell79_question_schema_source_contexts = (
    cell79_source_contexts(
        CELL79_QUESTION_SCHEMA_SEARCH_TERMS,
        context_before=15,
        context_after=25,
    )
)


# Runtime fields that look like actual future question-output
# fields, not lifecycle/status metadata.
CELL79_QUESTION_OUTPUT_FIELD_CANDIDATES = {
    "question",
    "question_text",
    "question_draft",
    "question_draft_text",
    "generated_question",
    "generated_question_text",
}


cell79_runtime_question_output_fields = sorted(
    {
        field_name

        for packet
        in question_drafting_pilot_packets

        for field_name
        in packet.keys()

        if (
            field_name.lower()
            in
            CELL79_QUESTION_OUTPUT_FIELD_CANDIDATES
        )
    }
)


# ============================================================
# 6) QUESTION ID SCHEMA evidence
# ============================================================

CELL79_QUESTION_ID_SEARCH_TERMS = [
    "question_id",
    "question id",
    "question_draft_id",
    "question draft id",
    "question-pilot",
    "pilot_case_id",
    "build_pilot_case_id",
    "build_question",
]


cell79_question_id_source_contexts = (
    cell79_source_contexts(
        CELL79_QUESTION_ID_SEARCH_TERMS,
        context_before=20,
        context_after=35,
    )
)


cell79_existing_question_pilot_ids = sorted(
    [
        packet[
            "pilot_case_id"
        ]

        for packet
        in question_drafting_pilot_packets
    ]
)


cell79_existing_question_pilot_prefix_counts = dict(
    Counter(
        pilot_id.split(
            ":",
            1,
        )[
            0
        ]

        if (
            ":"
            in
            pilot_id
        )

        else pilot_id

        for pilot_id
        in cell79_existing_question_pilot_ids
    )
)


# ============================================================
# 7) ACCEPTANCE RUBRIC evidence
# ============================================================

CELL79_RUBRIC_SEARCH_TERMS = [
    "acceptance",
    "rubric",
    "review",
    "validation",
    "validate",
    "groundability",
    "answer leakage",
    "internal ids",
    "external knowledge",
    "unsupported attribution",
    "locator fabrication",
    "required_dimensions",
]


cell79_rubric_source_contexts = (
    cell79_source_contexts(
        CELL79_RUBRIC_SEARCH_TERMS,
        context_before=15,
        context_after=25,
    )
)


# ============================================================
# 8) Runtime semantic dimensions available for future rubric
# ============================================================

cell79_required_dimension_counts = dict(
    Counter(
        dimension

        for packet
        in question_drafting_pilot_packets

        for dimension
        in packet.get(
            "required_dimensions",
            []
        )
    )
)


cell79_question_archetype_counts = dict(
    Counter(
        packet.get(
            "question_archetype"
        )

        for packet
        in question_drafting_pilot_packets
    )
)


cell79_task_family_counts = dict(
    Counter(
        packet.get(
            "task_family"
        )

        for packet
        in question_drafting_pilot_packets
    )
)


cell79_guardrail_contract = (
    cell79_clone(
        cell78_frozen_guardrails
    )
)


# ============================================================
# 9) Detect actual named schemas / builders / contracts
# ============================================================

CELL79_DEFINITION_TERMS = (
    "question",
    "rubric",
    "accept",
    "review",
    "validate",
)


cell79_related_function_definitions = []


for code_cell in (
    cell49_code_cells
):

    source_lines = (
        code_cell[
            "source"
        ].splitlines()
    )


    for line_index, line in enumerate(
        source_lines
    ):

        stripped = (
            line.strip()
        )


        if not (
            stripped.startswith(
                "def "
            )
        ):

            continue


        if not any(
            term
            in
            stripped.lower()

            for term
            in CELL79_DEFINITION_TERMS
        ):

            continue


        cell79_related_function_definitions.append(
            {
                "cell_index":
                    code_cell[
                        "cell_index"
                    ],

                "line_index":
                    line_index,

                "definition":
                    stripped,
            }
        )


# ============================================================
# 10) Resolve each design item's evidence status
#
# We only mark SOURCE_DEFINED if source/runtime contains
# concrete evidence for the design object itself.
#
# Lifecycle mentions alone do NOT count as definition.
# ============================================================

cell79_authoring_mode_status = (
    "SOURCE_EVIDENCE_PRESENT_REQUIRES_INTERPRETATION"

    if (
        len(
            cell79_authoring_mode_source_contexts
        )
        >
        0
    )

    else
    "NO_SOURCE_EVIDENCE_DESIGN_REQUIRED"
)


cell79_question_text_schema_status = (
    "RUNTIME_OUTPUT_SCHEMA_ALREADY_PRESENT"

    if (
        len(
            cell79_runtime_question_output_fields
        )
        >
        0
    )

    else
    "NO_RUNTIME_QUESTION_OUTPUT_SCHEMA_DESIGN_REQUIRED"
)


cell79_question_id_schema_status = (
    "PILOT_ID_SCHEMA_PRESENT_FINAL_QUESTION_ID_SCHEMA_UNRESOLVED"

    if (
        len(
            cell79_existing_question_pilot_ids
        )
        ==
        23

        and

        cell79_existing_question_pilot_prefix_counts
        ==
        {
            "question-pilot":
                23
        }
    )

    else
    "QUESTION_ID_LINEAGE_REQUIRES_FURTHER_REVIEW"
)


cell79_acceptance_rubric_status = (
    "VALIDATION_EVIDENCE_PRESENT_FORMAL_RUBRIC_NOT_YET_DEFINED"

    if (
        len(
            cell79_rubric_source_contexts
        )
        >
        0

        and

        len(
            cell79_required_dimension_counts
        )
        >
        0

        and

        isinstance(
            cell79_guardrail_contract,
            dict,
        )

        and

        len(
            cell79_guardrail_contract
        )
        >
        0
    )

    else
    "ACCEPTANCE_RUBRIC_EVIDENCE_INSUFFICIENT"
)


# ============================================================
# 11) Discovery failures
# ============================================================

cell79_discovery_failures = []


if (
    len(
        cell79_existing_question_pilot_ids
    )
    !=
    23
):

    cell79_discovery_failures.append(
        {
            "reason":
                "QUESTION_PILOT_ID_COUNT_NOT_23",

            "actual":
                len(
                    cell79_existing_question_pilot_ids
                ),
        }
    )


if (
    len(
        set(
            cell79_existing_question_pilot_ids
        )
    )
    !=
    23
):

    cell79_discovery_failures.append(
        "QUESTION_PILOT_IDS_NOT_UNIQUE"
    )


if (
    not isinstance(
        cell79_guardrail_contract,
        dict,
    )

    or

    len(
        cell79_guardrail_contract
    )
    ==
    0
):

    cell79_discovery_failures.append(
        "FROZEN_GUARDRAIL_CONTRACT_MISSING"
    )


if (
    len(
        cell79_required_dimension_counts
    )
    ==
    0
):

    cell79_discovery_failures.append(
        "REQUIRED_DIMENSION_EVIDENCE_MISSING"
    )


# ============================================================
# 12) Freeze design-evidence resolution
# ============================================================

PILOT_QUESTION_DRAFTING_DESIGN_EVIDENCE_V1_DRAFT = {
    "version":
        "pilot_question_drafting_design_evidence_v1_draft",

    "status":
        (
            "FOUR_DESIGN_ITEMS_EVIDENCE_RESOLVED_"
            "NO_GENERATION"

            if (
                len(
                    cell79_prerequisite_failures
                )
                ==
                0

                and

                len(
                    cell79_discovery_failures
                )
                ==
                0
            )

            else
            "FOUR_DESIGN_ITEMS_EVIDENCE_DISCOVERY_FAILED"
        ),

    "authoring_mode": {
        "status":
            cell79_authoring_mode_status,

        "source_context_count":
            len(
                cell79_authoring_mode_source_contexts
            ),

        "signal_counts":
            cell79_clone(
                cell79_authoring_mode_exact_signal_counts
            ),
    },

    "question_text_schema": {
        "status":
            cell79_question_text_schema_status,

        "runtime_output_fields":
            list(
                cell79_runtime_question_output_fields
            ),

        "source_context_count":
            len(
                cell79_question_schema_source_contexts
            ),
    },

    "question_id_schema": {
        "status":
            cell79_question_id_schema_status,

        "existing_pilot_id_count":
            len(
                cell79_existing_question_pilot_ids
            ),

        "prefix_counts":
            cell79_clone(
                cell79_existing_question_pilot_prefix_counts
            ),

        "source_context_count":
            len(
                cell79_question_id_source_contexts
            ),
    },

    "acceptance_rubric": {
        "status":
            cell79_acceptance_rubric_status,

        "source_context_count":
            len(
                cell79_rubric_source_contexts
            ),

        "required_dimension_counts":
            cell79_clone(
                cell79_required_dimension_counts
            ),

        "guardrails":
            cell79_clone(
                cell79_guardrail_contract
            ),
    },

    "question_generation":
        "NOT_STARTED",

    "expected_answer_generation":
        "NOT_STARTED",

    "citation_adjudication":
        "NOT_STARTED",

    "active_state_mutation":
        "NONE",

    "artifact_write":
        "NONE",
}


# ============================================================
# 13) Determine next action
# ============================================================

cell79_next_step = (
    "DESIGN_AND_VALIDATE_FOUR_PILOT_DRAFTING_COMPONENTS"
)


# ============================================================
# 14) Active-state snapshot AFTER
# ============================================================

cell79_active_snapshot_after = {
    "canonical":
        cell79_sha(
            {
                "golden_case_records":
                    golden_case_records,

                "case_groundability_precheck":
                    case_groundability_precheck,

                "case_task_design":
                    case_task_design,

                "task_draft_plan_rows":
                    task_draft_plan_rows,

                "target_evidence_records":
                    target_evidence_records,

                "pilot_selection_rows":
                    pilot_selection_rows,

                "golden_case_records_by_id":
                    golden_case_records_by_id,

                "task_draft_plans":
                    task_draft_plans,

                "target_evidence_by_case_id":
                    target_evidence_by_case_id,

                "target_evidence_by_task_draft_id":
                    target_evidence_by_task_draft_id,
            }
        ),

    "pilot_runtime":
        cell79_sha(
            {
                "question_drafting_pilot_packets":
                    question_drafting_pilot_packets,

                "question_drafting_pilot_by_id":
                    question_drafting_pilot_by_id,
            }
        ),
}


# ============================================================
# 15) Gates
# ============================================================

cell79_gates = {
    "cell78_boundary_contract_is_pass":
        (
            len(
                cell79_prerequisite_failures
            )
            ==
            0
        ),

    "authoring_mode_source_evidence_is_resolved_without_guessing":
        (
            cell79_authoring_mode_status
            in
            {
                "SOURCE_EVIDENCE_PRESENT_REQUIRES_INTERPRETATION",
                "NO_SOURCE_EVIDENCE_DESIGN_REQUIRED",
            }
        ),

    "question_text_schema_evidence_is_resolved":
        (
            cell79_question_text_schema_status
            in
            {
                "RUNTIME_OUTPUT_SCHEMA_ALREADY_PRESENT",
                "NO_RUNTIME_QUESTION_OUTPUT_SCHEMA_DESIGN_REQUIRED",
            }
        ),

    "question_id_lineage_is_resolved_to_current_23_question_pilots":
        (
            cell79_question_id_schema_status
            ==
            "PILOT_ID_SCHEMA_PRESENT_FINAL_QUESTION_ID_SCHEMA_UNRESOLVED"
        ),

    "acceptance_rubric_evidence_is_resolved_from_dimensions_and_guardrails":
        (
            cell79_acceptance_rubric_status
            ==
            "VALIDATION_EVIDENCE_PRESENT_FORMAL_RUBRIC_NOT_YET_DEFINED"
        ),

    "all_23_question_pilot_ids_are_unique":
        (
            len(
                set(
                    cell79_existing_question_pilot_ids
                )
            )
            ==
            23
        ),

    "runtime_guardrails_are_available_for_future_rubric":
        (
            isinstance(
                cell79_guardrail_contract,
                dict,
            )

            and

            len(
                cell79_guardrail_contract
            )
            >
            0
        ),

    "required_evaluation_dimensions_are_available_for_future_rubric":
        (
            len(
                cell79_required_dimension_counts
            )
            >
            0
        ),

    "design_evidence_discovery_has_zero_failures":
        (
            len(
                cell79_discovery_failures
            )
            ==
            0
        ),

    "four_design_items_are_resolved_as_evidence_not_yet_as_generation_contract":
        (
            PILOT_QUESTION_DRAFTING_DESIGN_EVIDENCE_V1_DRAFT[
                "status"
            ]
            ==
            "FOUR_DESIGN_ITEMS_EVIDENCE_RESOLVED_"
            "NO_GENERATION"
        ),

    "next_step_is_design_and_validation_not_generation":
        (
            cell79_next_step
            ==
            "DESIGN_AND_VALIDATE_FOUR_PILOT_DRAFTING_COMPONENTS"
        ),

    "active_state_remains_byte_stable":
        (
            cell79_active_snapshot_before
            ==
            cell79_active_snapshot_after
        ),

    "question_generation_not_started":
        (
            PILOT_QUESTION_DRAFTING_DESIGN_EVIDENCE_V1_DRAFT[
                "question_generation"
            ]
            ==
            "NOT_STARTED"
        ),

    "expected_answer_generation_not_started":
        (
            PILOT_QUESTION_DRAFTING_DESIGN_EVIDENCE_V1_DRAFT[
                "expected_answer_generation"
            ]
            ==
            "NOT_STARTED"
        ),

    "citation_adjudication_not_started":
        (
            PILOT_QUESTION_DRAFTING_DESIGN_EVIDENCE_V1_DRAFT[
                "citation_adjudication"
            ]
            ==
            "NOT_STARTED"
        ),

    "no_artifact_write":
        (
            PILOT_QUESTION_DRAFTING_DESIGN_EVIDENCE_V1_DRAFT[
                "artifact_write"
            ]
            ==
            "NONE"
        ),
}


cell_79_pass = all(
    cell79_gates.values()
)


# ============================================================
# 16) Output
# ============================================================

print(
    "=== PILOT QUESTION DRAFTING — FOUR DESIGN ITEMS EVIDENCE ==="
)


print(
    "\n=== PREREQUISITES ==="
)


print(
    "Failures:",
    cell79_prerequisite_failures
)


print(
    "\n=== 1) AUTHORING MODE EVIDENCE ==="
)


print(
    "Status:",
    cell79_authoring_mode_status
)


print(
    "Signal counts:",
    cell79_authoring_mode_exact_signal_counts
)


print(
    "Context count:",
    len(
        cell79_authoring_mode_source_contexts
    )
)


for row in (
    cell79_authoring_mode_source_contexts[
        :12
    ]
):

    print(
        "\nCELL",
        row[
            "cell_index"
        ],
        "LINE",
        row[
            "line_index"
        ],
        "MATCH=",
        row[
            "matched_terms"
        ]
    )


    print(
        row[
            "snippet"
        ]
    )


print(
    "\n=== 2) QUESTION TEXT SCHEMA EVIDENCE ==="
)


print(
    "Status:",
    cell79_question_text_schema_status
)


print(
    "Runtime output fields:",
    cell79_runtime_question_output_fields
)


print(
    "Source context count:",
    len(
        cell79_question_schema_source_contexts
    )
)


for row in (
    cell79_question_schema_source_contexts[
        :12
    ]
):

    print(
        "\nCELL",
        row[
            "cell_index"
        ],
        "LINE",
        row[
            "line_index"
        ],
        "MATCH=",
        row[
            "matched_terms"
        ]
    )


    print(
        row[
            "snippet"
        ]
    )


print(
    "\n=== 3) QUESTION ID SCHEMA EVIDENCE ==="
)


print(
    "Status:",
    cell79_question_id_schema_status
)


print(
    "Pilot prefix counts:",
    cell79_existing_question_pilot_prefix_counts
)


print(
    "Pilot ID count:",
    len(
        cell79_existing_question_pilot_ids
    )
)


print(
    "Source context count:",
    len(
        cell79_question_id_source_contexts
    )
)


for row in (
    cell79_question_id_source_contexts[
        :12
    ]
):

    print(
        "\nCELL",
        row[
            "cell_index"
        ],
        "LINE",
        row[
            "line_index"
        ],
        "MATCH=",
        row[
            "matched_terms"
        ]
    )


    print(
        row[
            "snippet"
        ]
    )


print(
    "\n=== 4) ACCEPTANCE RUBRIC EVIDENCE ==="
)


print(
    "Status:",
    cell79_acceptance_rubric_status
)


print(
    "Required dimensions:",
    cell79_required_dimension_counts
)


print(
    "Guardrails:",
    cell79_guardrail_contract
)


print(
    "Source context count:",
    len(
        cell79_rubric_source_contexts
    )
)


for row in (
    cell79_rubric_source_contexts[
        :12
    ]
):

    print(
        "\nCELL",
        row[
            "cell_index"
        ],
        "LINE",
        row[
            "line_index"
        ],
        "MATCH=",
        row[
            "matched_terms"
        ]
    )


    print(
        row[
            "snippet"
        ]
    )


print(
    "\n=== RELATED FUNCTION DEFINITIONS ==="
)


print(
    "Count:",
    len(
        cell79_related_function_definitions
    )
)


for row in (
    cell79_related_function_definitions
):

    print(
        "- CELL",
        row[
            "cell_index"
        ],
        "LINE",
        row[
            "line_index"
        ],
        ":",
        row[
            "definition"
        ]
    )


print(
    "\n=== DISCOVERY FAILURES ==="
)


print(
    cell79_discovery_failures
)


print(
    "\nSTATUS:",
    PILOT_QUESTION_DRAFTING_DESIGN_EVIDENCE_V1_DRAFT[
        "status"
    ]
)


print(
    "NEXT STEP:",
    cell79_next_step
)


print(
    "\n=== CODE CELL 79 GATES ==="
)


for gate_name, passed in (
    cell79_gates.items()
):

    print(
        gate_name,
        ":",
        "PASS"
        if passed
        else "FAIL"
    )


print(
    "\nFAILED GATES:",
    [
        gate_name

        for gate_name, passed
        in cell79_gates.items()

        if not passed
    ]
)


print(
    "\nCODE CELL 79 GATE:",
    "PASS"
    if cell_79_pass
    else "FAIL"
)


assert cell_79_pass, (
    "The four Pilot Question Drafting design items could "
    "not be resolved from source/runtime evidence. "
    "Do not design or generate questions. Repair this SAME "
    "Code Cell 79 using the printed authoring-mode, schema, "
    "ID-lineage and acceptance-rubric diagnostics."
)
```

<h3>Pilot Question Drafting — Four-Component Design Freeze</h3>

<p>
Cell 79 نشان داد که Source Notebook اصول و Guardrailهای Question Drafting را دارد،
اما چهار جزء اجرایی زیر هنوز Contract کامل و مستقلی ندارند:
</p>

<ul>
<li>Authoring Mode</li>
<li>Question Draft Record Schema</li>
<li>Stable Question-Draft ID Schema</li>
<li>Question Acceptance Rubric</li>
</ul>

<p>
این Cell این چهار جزء را به‌عنوان <strong>طراحی جدید NB12</strong> تعریف می‌کند.
هیچ‌یک به‌اشتباه به‌عنوان Contract تاریخی یا Source-defined معرفی نمی‌شوند.
</p>

<p>
Authoring Mode برای Pilot به‌صورت
<strong>MODEL_DRAFT_WITH_HUMAN_ADJUDICATION</strong>
تعریف می‌شود:
Model می‌تواند Draft اولیه را فقط از Evidence Packet بسازد،
اما هیچ Draftی بدون Validation و Human Adjudication به Golden Question پذیرفته نمی‌شود.
</p>

<p>
Question Draft ID deterministic است و از Pilot identity و hash متن Draft ساخته می‌شود.
بنابراین بازنویسی سؤال به‌صورت عمدی ID جدید می‌سازد و lineage نسخه‌های Draft حفظ می‌شود.
</p>

<p>
Acceptance Rubric ترکیبی از Hard Guardrail Gates و Semantic Validation Gates است.
تا زمانی که همه Hard Gateها PASS نشوند، Expected Answer و Citation Validation شروع نمی‌شوند.
</p>

<p>
این Cell فقط Contract را Freeze می‌کند.
هیچ متن سؤال، Expected Answer، Citation یا Artifact تولید نمی‌شود.
</p>

```python
# Code Cell 80 — REPAIRED V2
#
# ============================================================
# Pilot Question Drafting — Four-Component Design Freeze
#
# REPAIR:
#   Cell80 must not depend on transient Cell79 runtime names.
#
#   Required evidence is independently reconstructed from:
#       - current 23 Pilot packets
#       - current Active task/target state
#       - frozen packet Guardrails
#
#   If Cell79 objects are still available, they are inspected
#   only as optional lineage evidence.
#
# NO question generation.
# NO answer generation.
# NO citation adjudication.
# NO active mutation.
# NO artifact write.
# ============================================================


# ============================================================
# 1) Helpers
# ============================================================

def cell80_sha(value):

    return hashlib.sha256(
        stable_json_key(
            value
        ).encode(
            "utf-8"
        )
    ).hexdigest()


def cell80_runtime_get(
    variable_name,
    default=None,
):

    return globals().get(
        variable_name,
        default,
    )


# ============================================================
# 2) Resolve required runtime objects WITHOUT Cell79 dependency
# ============================================================

cell80_packets = (
    cell80_runtime_get(
        "question_drafting_pilot_packets"
    )
)


cell80_packet_index = (
    cell80_runtime_get(
        "question_drafting_pilot_by_id"
    )
)


cell80_required_runtime_names = [
    "question_drafting_pilot_packets",
    "question_drafting_pilot_by_id",
    "golden_case_records",
    "case_groundability_precheck",
    "case_task_design",
    "task_draft_plan_rows",
    "target_evidence_records",
    "pilot_selection_rows",
    "golden_case_records_by_id",
    "task_draft_plans",
    "target_evidence_by_case_id",
    "target_evidence_by_task_draft_id",
]


cell80_missing_runtime_objects = [
    variable_name

    for variable_name
    in cell80_required_runtime_names

    if (
        variable_name
        not in
        globals()
    )
]


cell80_prerequisite_failures = []


if (
    cell80_missing_runtime_objects
):

    cell80_prerequisite_failures.append(
        {
            "reason":
                "REQUIRED_RUNTIME_OBJECTS_MISSING",

            "objects":
                cell80_missing_runtime_objects,
        }
    )


if not (
    isinstance(
        cell80_packets,
        list,
    )

    and

    len(
        cell80_packets
    )
    ==
    23
):

    cell80_prerequisite_failures.append(
        {
            "reason":
                "CURRENT_PILOT_PACKET_COUNT_NOT_23",

            "actual":
                (
                    len(
                        cell80_packets
                    )
                    if isinstance(
                        cell80_packets,
                        list,
                    )
                    else None
                ),
        }
    )


if not (
    isinstance(
        cell80_packet_index,
        dict,
    )

    and

    len(
        cell80_packet_index
    )
    ==
    23
):

    cell80_prerequisite_failures.append(
        {
            "reason":
                "CURRENT_PILOT_PACKET_INDEX_COUNT_NOT_23",

            "actual":
                (
                    len(
                        cell80_packet_index
                    )
                    if isinstance(
                        cell80_packet_index,
                        dict,
                    )
                    else None
                ),
        }
    )


# ============================================================
# 3) Optional Cell79 lineage evidence
#
# Absence is NOT a failure.
# This makes Cell80 robust to a restarted/reordered kernel.
# ============================================================

cell80_cell79_pass_runtime = (
    cell80_runtime_get(
        "cell_79_pass"
    )
)


cell80_cell79_contract_runtime = (
    cell80_runtime_get(
        "PILOT_QUESTION_DRAFTING_DESIGN_EVIDENCE_V1_DRAFT"
    )
)


cell80_cell79_lineage_status = (
    "AVAILABLE_AND_PASS"

    if (
        cell80_cell79_pass_runtime
        is True
    )

    else
    (
        "AVAILABLE_BUT_NOT_PASS"

        if (
            cell80_cell79_pass_runtime
            is not None
        )

        else
        "NOT_AVAILABLE_EVIDENCE_REDERIVED_IN_CELL80"
    )
)


if (
    cell80_cell79_pass_runtime
    is False
):

    cell80_prerequisite_failures.append(
        "CELL79_PRESENT_BUT_NOT_PASS"
    )


if isinstance(
    cell80_cell79_contract_runtime,
    dict,
):

    cell80_cell79_contract_status = (
        cell80_cell79_contract_runtime.get(
            "status"
        )
    )


    if (
        cell80_cell79_contract_status
        !=
        "FOUR_DESIGN_ITEMS_EVIDENCE_RESOLVED_"
        "NO_GENERATION"
    ):

        cell80_prerequisite_failures.append(
            {
                "reason":
                    "CELL79_CONTRACT_PRESENT_WITH_UNEXPECTED_STATUS",

                "actual":
                    cell80_cell79_contract_status,
            }
        )


else:

    cell80_cell79_contract_status = (
        "NOT_AVAILABLE"
    )


# ============================================================
# 4) Snapshot ACTIVE state BEFORE
# ============================================================

cell80_active_state_object_names = [
    "golden_case_records",
    "case_groundability_precheck",
    "case_task_design",
    "task_draft_plan_rows",
    "target_evidence_records",
    "pilot_selection_rows",
    "golden_case_records_by_id",
    "task_draft_plans",
    "target_evidence_by_case_id",
    "target_evidence_by_task_draft_id",
]


cell80_active_snapshot_before = (
    cell80_sha(
        {
            variable_name:
                globals()[
                    variable_name
                ]

            for variable_name
            in cell80_active_state_object_names

            if (
                variable_name
                in
                globals()
            )
        }
    )
)


cell80_pilot_runtime_snapshot_before = (
    cell80_sha(
        {
            "question_drafting_pilot_packets":
                cell80_packets,

            "question_drafting_pilot_by_id":
                cell80_packet_index,
        }
    )
)


# ============================================================
# 5) Independently rederive design evidence from the
# current 23 Pilot packets
# ============================================================

cell80_required_dimension_counts = dict(
    Counter(
        dimension

        for packet
        in (
            cell80_packets
            or []
        )

        for dimension
        in packet.get(
            "required_dimensions",
            []
        )
    )
)


cell80_task_family_counts = dict(
    Counter(
        packet.get(
            "task_family"
        )

        for packet
        in (
            cell80_packets
            or []
        )
    )
)


cell80_question_archetype_counts = dict(
    Counter(
        packet.get(
            "question_archetype"
        )

        for packet
        in (
            cell80_packets
            or []
        )
    )
)


cell80_evidence_scope_counts = dict(
    Counter(
        packet.get(
            "evidence_scope"
        )

        for packet
        in (
            cell80_packets
            or []
        )
    )
)


# ============================================================
# 6) Independently recover exact runtime Guardrails
# ============================================================

cell80_guardrail_variants = {}


for packet in (
    cell80_packets
    or []
):

    guardrails = (
        packet.get(
            "guardrails"
        )
    )


    guardrail_hash = (
        cell80_sha(
            guardrails
        )
    )


    if (
        guardrail_hash
        not in
        cell80_guardrail_variants
    ):

        cell80_guardrail_variants[
            guardrail_hash
        ] = {
            "value":
                guardrails,

            "count":
                0,
        }


    cell80_guardrail_variants[
        guardrail_hash
    ][
        "count"
    ] += 1


cell80_guardrail_failures = []


if (
    len(
        cell80_guardrail_variants
    )
    !=
    1
):

    cell80_guardrail_failures.append(
        {
            "reason":
                "RUNTIME_GUARDRAILS_NOT_UNIFORM",

            "variant_count":
                len(
                    cell80_guardrail_variants
                ),
        }
    )


cell80_frozen_guardrails = (
    next(
        iter(
            cell80_guardrail_variants.values()
        )
    )[
        "value"
    ]

    if (
        len(
            cell80_guardrail_variants
        )
        ==
        1
    )

    else None
)


CELL80_EXPECTED_RUNTIME_GUARDRAILS = {
    "external_knowledge":
        "FORBIDDEN",

    "unknown_speaker_inference":
        "FORBIDDEN",

    "unsupported_attribution":
        "FORBIDDEN",

    "secondary_to_primary_promotion":
        "FORBIDDEN",

    "locator_fabrication":
        "FORBIDDEN",

    "answer_leakage_in_question":
        "FORBIDDEN",

    "internal_ids_in_question":
        "FORBIDDEN",
}


if (
    cell80_frozen_guardrails
    !=
    CELL80_EXPECTED_RUNTIME_GUARDRAILS
):

    cell80_guardrail_failures.append(
        {
            "reason":
                "RUNTIME_GUARDRAILS_UNEXPECTED",

            "expected":
                CELL80_EXPECTED_RUNTIME_GUARDRAILS,

            "actual":
                cell80_frozen_guardrails,
        }
    )


# ============================================================
# 7) Source-derived Question rules
#
# These were explicitly discovered from Cell51 source.
# ============================================================

CELL80_SOURCE_DERIVED_QUESTION_RULES = {
    "natural_user_question":
        True,

    "internal_document_id_leakage":
        "FORBIDDEN",

    "hash_or_internal_pipeline_metadata_leakage":
        "FORBIDDEN",

    "answer_leakage_in_question":
        "FORBIDDEN",

    "external_knowledge_dependency":
        "FORBIDDEN",

    "unsupported_attribution":
        "FORBIDDEN",

    "unsupported_authority_promotion":
        "FORBIDDEN",

    "unknown_speaker_inference":
        "FORBIDDEN",
}


CELL80_SOURCE_DERIVED_SEQUENCE = [
    "CASE_VALIDATED",
    "TASK_DESIGN_VALIDATED",
    "GROUNDABILITY_PRECHECK_PASS",
    "TARGET_EVIDENCE_SELECTION",
    "TARGET_EVIDENCE_VALIDATION",
    "QUESTION_DRAFT",
    "QUESTION_GROUNDABILITY_VALIDATION",
    "EXPECTED_ANSWER_CONSTRUCTION",
    "CITATION_TARGET_VALIDATION",
]


CELL80_SOURCE_DERIVED_CRITICAL_RULE = (
    "TARGET_EVIDENCE_MUST_BE_SELECTED_AND_VALIDATED_"
    "BEFORE_QUESTION_TEXT"
)


# ============================================================
# 8) Component 1 — Authoring Mode
#
# NEW NB12 DESIGN
# ============================================================

PILOT_QUESTION_AUTHORING_MODE_V1 = {
    "version":
        "pilot_question_authoring_mode_v1",

    "design_lineage":
        "NEWLY_DEFINED_IN_NB12_CELL80",

    "mode":
        "MODEL_DRAFT_WITH_HUMAN_ADJUDICATION",

    "model_role":
        (
            "CREATE_INITIAL_QUESTION_DRAFT_FROM_THE_"
            "CURRENT_PILOT_PACKET_ONLY"
        ),

    "human_role":
        (
            "ADJUDICATE_ACCEPT_REJECT_OR_REQUEST_REVISION_"
            "AFTER_MACHINE_VALIDATION"
        ),

    "automatic_acceptance":
        False,

    "allowed_external_knowledge":
        False,

    "allowed_evidence_source":
        "CURRENT_PILOT_PACKET_TARGET_TEXT_PARTS_ONLY",

    "pilot_scope":
        23,

    "full_population_generation":
        "BLOCKED_PENDING_PILOT_VALIDATION",
}


# ============================================================
# 9) Component 2 — Question Draft Record Schema
#
# NEW NB12 DESIGN
# ============================================================

QUESTION_DRAFT_RECORD_SCHEMA_V1 = {
    "schema_version":
        "question_draft_record_v1",

    "design_lineage":
        "NEWLY_DEFINED_IN_NB12_CELL80",

    "required_fields": [
        "question_draft_id",
        "question_draft_schema_version",
        "pilot_case_id",
        "golden_case_id",
        "task_draft_id",
        "target_evidence_id",
        "source_type",
        "stratum",
        "task_family",
        "question_archetype",
        "required_dimensions",
        "authority_expectation",
        "context_expectation",
        "evidence_scope",
        "question_text",
        "question_text_sha256",
        "authoring_mode",
        "draft_status",
        "machine_validation_status",
        "human_adjudication_status",
        "expected_answer_status",
        "citation_target_validation_status",
    ],

    "question_text_rules":
        dict(
            CELL80_SOURCE_DERIVED_QUESTION_RULES
        ),

    "initial_statuses": {
        "draft_status":
            "DRAFT_CREATED_PENDING_VALIDATION",

        "machine_validation_status":
            "NOT_STARTED",

        "human_adjudication_status":
            "NOT_STARTED",

        "expected_answer_status":
            "BLOCKED_PENDING_QUESTION_ACCEPTANCE",

        "citation_target_validation_status":
            "BLOCKED_PENDING_QUESTION_ACCEPTANCE",
    },

    "final_question_acceptance_required":
        True,

    "question_text_must_be_nonempty":
        True,

    "question_text_must_be_natural_language":
        True,

    "question_must_be_groundable_in_target_evidence":
        True,
}


# ============================================================
# 10) Component 3 — Stable Question-Draft ID
#
# NEW NB12 DESIGN
# ============================================================

QUESTION_DRAFT_ID_METHOD_VERSION = (
    "question_draft_id_method_v1"
)


def build_question_draft_id(
    pilot_case_id,
    question_text_sha256,
):

    payload = {
        "method_version":
            QUESTION_DRAFT_ID_METHOD_VERSION,

        "pilot_case_id":
            pilot_case_id,

        "question_text_sha256":
            question_text_sha256,
    }


    digest = hashlib.sha256(
        stable_json_key(
            payload
        ).encode(
            "utf-8"
        )
    ).hexdigest()


    return (
        "question-draft:"
        + digest[
            :32
        ]
    )


QUESTION_DRAFT_ID_CONTRACT_V1 = {
    "version":
        QUESTION_DRAFT_ID_METHOD_VERSION,

    "design_lineage":
        "NEWLY_DEFINED_IN_NB12_CELL80",

    "prefix":
        "question-draft:",

    "identity_inputs": [
        "pilot_case_id",
        "question_text_sha256",
    ],

    "hash":
        "SHA256",

    "digest_chars":
        32,

    "same_exact_draft_reproduces_same_id":
        True,

    "question_revision_creates_new_id":
        True,

    "pilot_case_identity_remains_parent_lineage":
        True,
}


# ============================================================
# 11) Independent deterministic ID validation
# ============================================================

CELL80_ID_TEST_PILOT = (
    "question-pilot:"
    "00000000000000000000000000000000"
)


CELL80_ID_TEST_TEXT = (
    "نمونه سؤال آزمایشی صرفاً برای آزمون تابع شناسه"
)


cell80_id_test_text_sha256 = (
    hashlib.sha256(
        CELL80_ID_TEST_TEXT.encode(
            "utf-8"
        )
    ).hexdigest()
)


cell80_id_test_a = (
    build_question_draft_id(
        pilot_case_id=(
            CELL80_ID_TEST_PILOT
        ),

        question_text_sha256=(
            cell80_id_test_text_sha256
        ),
    )
)


cell80_id_test_b = (
    build_question_draft_id(
        pilot_case_id=(
            CELL80_ID_TEST_PILOT
        ),

        question_text_sha256=(
            cell80_id_test_text_sha256
        ),
    )
)


cell80_id_test_revision_sha256 = (
    hashlib.sha256(
        (
            CELL80_ID_TEST_TEXT
            + " نسخه دوم"
        ).encode(
            "utf-8"
        )
    ).hexdigest()
)


cell80_id_test_revision = (
    build_question_draft_id(
        pilot_case_id=(
            CELL80_ID_TEST_PILOT
        ),

        question_text_sha256=(
            cell80_id_test_revision_sha256
        ),
    )
)


cell80_id_contract_failures = []


if (
    cell80_id_test_a
    !=
    cell80_id_test_b
):

    cell80_id_contract_failures.append(
        "QUESTION_DRAFT_ID_NOT_DETERMINISTIC"
    )


if (
    cell80_id_test_a
    ==
    cell80_id_test_revision
):

    cell80_id_contract_failures.append(
        "QUESTION_REVISION_DID_NOT_CHANGE_DRAFT_ID"
    )


if not (
    cell80_id_test_a.startswith(
        "question-draft:"
    )
):

    cell80_id_contract_failures.append(
        "QUESTION_DRAFT_ID_PREFIX_INVALID"
    )


# ============================================================
# 12) Component 4 — Acceptance Rubric
#
# NEW NB12 DESIGN
# ============================================================

QUESTION_DRAFT_ACCEPTANCE_RUBRIC_V1 = {
    "version":
        "question_draft_acceptance_rubric_v1",

    "design_lineage":
        "NEWLY_DEFINED_IN_NB12_CELL80",

    "decision_values": [
        "ACCEPT",
        "REVISE",
        "REJECT",
    ],

    "hard_gates": {
        "nonempty_question_text":
            True,

        "natural_user_question":
            True,

        "external_knowledge_dependency":
            "FORBIDDEN",

        "answer_leakage_in_question":
            "FORBIDDEN",

        "internal_document_id_leakage":
            "FORBIDDEN",

        "internal_pipeline_metadata_leakage":
            "FORBIDDEN",

        "unknown_speaker_inference":
            "FORBIDDEN",

        "unsupported_attribution":
            "FORBIDDEN",

        "unsupported_authority_promotion":
            "FORBIDDEN",

        "locator_fabrication":
            "FORBIDDEN",

        "question_must_be_groundable_in_target_text_parts":
            True,

        "target_evidence_must_already_be_validated":
            True,
    },

    "semantic_gates": {
        "task_family_alignment":
            "PASS_REQUIRED",

        "question_archetype_alignment":
            "PASS_REQUIRED",

        "evidence_scope_alignment":
            "PASS_REQUIRED",

        "authority_expectation_alignment":
            "PASS_REQUIRED",

        "context_expectation_alignment":
            "PASS_REQUIRED",

        "required_dimensions_are_testable_by_question":
            "PASS_REQUIRED",

        "question_is_not_trivial_metadata_repetition":
            "PASS_REQUIRED",

        "question_is_answerable_without_external_context":
            "PASS_REQUIRED",
    },

    "dimension_policy": {
        "packet_required_dimensions_are_authoritative":
            True,

        "global_dimension_universe":
            sorted(
                cell80_required_dimension_counts.keys()
            ),
    },

    "human_review_policy": {
        "required_before_acceptance":
            True,

        "allowed_decisions": [
            "ACCEPT",
            "REVISE",
            "REJECT",
        ],

        "ambiguous_groundability":
            "REVISE_OR_REJECT",

        "ambiguous_attribution":
            "REVISE_OR_REJECT",

        "ambiguous_authority":
            "REVISE_OR_REJECT",

        "ambiguous_evidence_boundary":
            "REVISE_OR_REJECT",
    },

    "acceptance_condition":
        (
            "ALL_HARD_GATES_PASS_AND_ALL_SEMANTIC_GATES_PASS_"
            "AND_HUMAN_DECISION_ACCEPT"
        ),

    "post_acceptance_unblocks": [
        "QUESTION_GROUNDABILITY_VALIDATION_COMPLETION",
        "EXPECTED_ANSWER_CONSTRUCTION",
        "CITATION_TARGET_VALIDATION",
    ],
}


# ============================================================
# 13) Exact Runtime Guardrail → Rubric mapping
# ============================================================

CELL80_RUNTIME_TO_RUBRIC_GUARDRAIL_MAP = {
    "external_knowledge":
        "external_knowledge_dependency",

    "unknown_speaker_inference":
        "unknown_speaker_inference",

    "unsupported_attribution":
        "unsupported_attribution",

    "secondary_to_primary_promotion":
        "unsupported_authority_promotion",

    "locator_fabrication":
        "locator_fabrication",

    "answer_leakage_in_question":
        "answer_leakage_in_question",

    "internal_ids_in_question":
        "internal_document_id_leakage",
}


cell80_rubric_consistency_failures = []


cell80_hard_gates = (
    QUESTION_DRAFT_ACCEPTANCE_RUBRIC_V1[
        "hard_gates"
    ]
)


if isinstance(
    cell80_frozen_guardrails,
    dict,
):

    if (
        set(
            CELL80_RUNTIME_TO_RUBRIC_GUARDRAIL_MAP.keys()
        )
        !=
        set(
            cell80_frozen_guardrails.keys()
        )
    ):

        cell80_rubric_consistency_failures.append(
            {
                "reason":
                    "RUNTIME_GUARDRAIL_MAPPING_COVERAGE_NOT_EXACT",

                "runtime_keys":
                    sorted(
                        cell80_frozen_guardrails.keys()
                    ),

                "mapping_keys":
                    sorted(
                        CELL80_RUNTIME_TO_RUBRIC_GUARDRAIL_MAP.keys()
                    ),
            }
        )


    for runtime_guardrail, runtime_value in (
        cell80_frozen_guardrails.items()
    ):

        rubric_field = (
            CELL80_RUNTIME_TO_RUBRIC_GUARDRAIL_MAP.get(
                runtime_guardrail
            )
        )


        if (
            rubric_field
            not in
            cell80_hard_gates
        ):

            cell80_rubric_consistency_failures.append(
                {
                    "reason":
                        "MAPPED_RUBRIC_FIELD_MISSING",

                    "runtime_guardrail":
                        runtime_guardrail,

                    "rubric_field":
                        rubric_field,
                }
            )

            continue


        if (
            runtime_value
            ==
            "FORBIDDEN"

            and

            cell80_hard_gates[
                rubric_field
            ]
            !=
            "FORBIDDEN"
        ):

            cell80_rubric_consistency_failures.append(
                {
                    "reason":
                        "RUNTIME_TO_RUBRIC_VALUE_MISMATCH",

                    "runtime_guardrail":
                        runtime_guardrail,

                    "runtime_value":
                        runtime_value,

                    "rubric_field":
                        rubric_field,

                    "rubric_value":
                        cell80_hard_gates[
                            rubric_field
                        ],
                }
            )


# ============================================================
# 14) Source-derived rules → Rubric mapping
# ============================================================

CELL80_SOURCE_RULE_TO_RUBRIC_MAP = {
    "natural_user_question":
        "natural_user_question",

    "internal_document_id_leakage":
        "internal_document_id_leakage",

    "hash_or_internal_pipeline_metadata_leakage":
        "internal_pipeline_metadata_leakage",

    "answer_leakage_in_question":
        "answer_leakage_in_question",

    "external_knowledge_dependency":
        "external_knowledge_dependency",

    "unsupported_attribution":
        "unsupported_attribution",

    "unsupported_authority_promotion":
        "unsupported_authority_promotion",

    "unknown_speaker_inference":
        "unknown_speaker_inference",
}


cell80_source_rule_consistency_failures = []


for source_rule, source_value in (
    CELL80_SOURCE_DERIVED_QUESTION_RULES.items()
):

    rubric_field = (
        CELL80_SOURCE_RULE_TO_RUBRIC_MAP.get(
            source_rule
        )
    )


    if (
        rubric_field
        not in
        cell80_hard_gates
    ):

        cell80_source_rule_consistency_failures.append(
            {
                "reason":
                    "SOURCE_RULE_NOT_IN_RUBRIC",

                "source_rule":
                    source_rule,

                "rubric_field":
                    rubric_field,
            }
        )

        continue


    rubric_value = (
        cell80_hard_gates[
            rubric_field
        ]
    )


    if (
        source_value
        is True

        and

        rubric_value
        is not True
    ):

        cell80_source_rule_consistency_failures.append(
            {
                "reason":
                    "SOURCE_TRUE_RULE_VALUE_MISMATCH",

                "source_rule":
                    source_rule,

                "source_value":
                    source_value,

                "rubric_value":
                    rubric_value,
            }
        )


    elif (
        source_value
        ==
        "FORBIDDEN"

        and

        rubric_value
        !=
        "FORBIDDEN"
    ):

        cell80_source_rule_consistency_failures.append(
            {
                "reason":
                    "SOURCE_FORBIDDEN_RULE_VALUE_MISMATCH",

                "source_rule":
                    source_rule,

                "source_value":
                    source_value,

                "rubric_value":
                    rubric_value,
            }
        )


# ============================================================
# 15) Validate all 23 current Pilot packets
# ============================================================

CELL80_REQUIRED_PACKET_FIELDS = [
    "pilot_case_id",
    "golden_case_id",
    "task_draft_id",
    "target_evidence_id",
    "source_type",
    "stratum",
    "task_family",
    "question_archetype",
    "required_dimensions",
    "authority_expectation",
    "context_expectation",
    "evidence_scope",
    "target_text_parts",
    "guardrails",
    "question_draft_status",
    "question_groundability_status",
    "expected_answer_status",
    "citation_target_validation_status",
]


CELL80_REQUIRED_PREGEN_STATUSES = {
    "question_draft_status":
        "NOT_STARTED",

    "question_groundability_status":
        "NOT_STARTED",

    "expected_answer_status":
        "NOT_STARTED",

    "citation_target_validation_status":
        "NOT_STARTED",
}


cell80_packet_contract_failures = []


for packet in (
    cell80_packets
    or []
):

    pilot_case_id = (
        packet.get(
            "pilot_case_id"
        )
    )


    missing_fields = [
        field_name

        for field_name
        in CELL80_REQUIRED_PACKET_FIELDS

        if (
            field_name
            not in
            packet
        )
    ]


    if (
        missing_fields
    ):

        cell80_packet_contract_failures.append(
            {
                "pilot_case_id":
                    pilot_case_id,

                "reason":
                    "REQUIRED_FIELDS_MISSING",

                "fields":
                    missing_fields,
            }
        )


    for field_name, expected_value in (
        CELL80_REQUIRED_PREGEN_STATUSES.items()
    ):

        if (
            packet.get(
                field_name
            )
            !=
            expected_value
        ):

            cell80_packet_contract_failures.append(
                {
                    "pilot_case_id":
                        pilot_case_id,

                    "reason":
                        "PREGEN_STATUS_MISMATCH",

                    "field":
                        field_name,

                    "expected":
                        expected_value,

                    "actual":
                        packet.get(
                            field_name
                        ),
                }
            )


    target_parts = (
        packet.get(
            "target_text_parts"
        )
    )


    if not (
        isinstance(
            target_parts,
            list,
        )

        and

        len(
            target_parts
        )
        >
        0
    ):

        cell80_packet_contract_failures.append(
            {
                "pilot_case_id":
                    pilot_case_id,

                "reason":
                    "TARGET_TEXT_PARTS_EMPTY",
            }
        )


    for semantic_field in [
        "task_family",
        "question_archetype",
        "required_dimensions",
        "authority_expectation",
        "context_expectation",
        "evidence_scope",
    ]:

        semantic_value = (
            packet.get(
                semantic_field
            )
        )


        if (
            semantic_value is None

            or

            semantic_value == ""

            or

            semantic_value == []
        ):

            cell80_packet_contract_failures.append(
                {
                    "pilot_case_id":
                        pilot_case_id,

                    "reason":
                        "SEMANTIC_FIELD_EMPTY",

                    "field":
                        semantic_field,
                }
            )


# ============================================================
# 16) Validate current Active Task/Target readiness for 23
# ============================================================

cell80_pilot_task_ids = {
    packet[
        "task_draft_id"
    ]

    for packet
    in (
        cell80_packets
        or []
    )

    if (
        "task_draft_id"
        in
        packet
    )
}


cell80_pilot_target_ids = {
    packet[
        "target_evidence_id"
    ]

    for packet
    in (
        cell80_packets
        or []
    )

    if (
        "target_evidence_id"
        in
        packet
    )
}


cell80_resolved_pilot_tasks = [
    row

    for row
    in (
        cell80_runtime_get(
            "task_draft_plan_rows",
            [],
        )
    )

    if (
        row.get(
            "task_draft_id"
        )
        in
        cell80_pilot_task_ids
    )
]


cell80_resolved_pilot_targets = [
    row

    for row
    in (
        cell80_runtime_get(
            "target_evidence_records",
            [],
        )
    )

    if (
        row.get(
            "target_evidence_id"
        )
        in
        cell80_pilot_target_ids
    )
]


cell80_active_readiness_failures = []


if (
    len(
        cell80_resolved_pilot_tasks
    )
    !=
    23
):

    cell80_active_readiness_failures.append(
        {
            "reason":
                "ACTIVE_PILOT_TASK_COUNT_NOT_23",

            "actual":
                len(
                    cell80_resolved_pilot_tasks
                ),
        }
    )


if (
    len(
        cell80_resolved_pilot_targets
    )
    !=
    23
):

    cell80_active_readiness_failures.append(
        {
            "reason":
                "ACTIVE_PILOT_TARGET_COUNT_NOT_23",

            "actual":
                len(
                    cell80_resolved_pilot_targets
                ),
        }
    )


cell80_task_construction_counts = dict(
    Counter(
        row.get(
            "construction_eligibility",
            "__MISSING__",
        )

        for row
        in cell80_resolved_pilot_tasks
    )
)


cell80_task_target_status_counts = dict(
    Counter(
        row.get(
            "target_evidence_status",
            "__MISSING__",
        )

        for row
        in cell80_resolved_pilot_tasks
    )
)


if (
    cell80_task_construction_counts
    !=
    {
        "READY_FOR_QUESTION_DRAFT":
            23
    }
):

    cell80_active_readiness_failures.append(
        {
            "reason":
                "PILOT_TASKS_NOT_READY_FOR_QUESTION_DRAFT",

            "actual":
                cell80_task_construction_counts,
        }
    )


if (
    cell80_task_target_status_counts
    !=
    {
        "SELECTED_AND_VALIDATED":
            23
    }
):

    cell80_active_readiness_failures.append(
        {
            "reason":
                "PILOT_TARGETS_NOT_SELECTED_AND_VALIDATED",

            "actual":
                cell80_task_target_status_counts,
        }
    )


# ============================================================
# 17) Complete Four-Component Contract
# ============================================================

PILOT_QUESTION_DRAFTING_FOUR_COMPONENT_CONTRACT_V1 = {
    "version":
        "pilot_question_drafting_four_component_contract_v1",

    "design_lineage":
        "NEWLY_DEFINED_IN_NB12_CELL80",

    "cell79_runtime_lineage":
        cell80_cell79_lineage_status,

    "authoring_mode":
        PILOT_QUESTION_AUTHORING_MODE_V1,

    "question_draft_record_schema":
        QUESTION_DRAFT_RECORD_SCHEMA_V1,

    "question_draft_id_contract":
        QUESTION_DRAFT_ID_CONTRACT_V1,

    "acceptance_rubric":
        QUESTION_DRAFT_ACCEPTANCE_RUBRIC_V1,

    "runtime_to_rubric_guardrail_map":
        dict(
            CELL80_RUNTIME_TO_RUBRIC_GUARDRAIL_MAP
        ),

    "workflow_sequence":
        list(
            CELL80_SOURCE_DERIVED_SEQUENCE
        ),

    "critical_source_rule":
        CELL80_SOURCE_DERIVED_CRITICAL_RULE,

    "required_dimension_counts":
        dict(
            cell80_required_dimension_counts
        ),

    "pilot_count":
        23,

    "full_population_count":
        572,

    "full_population_generation":
        "BLOCKED_PENDING_23_PILOT_ACCEPTANCE",

    "question_generation":
        "NOT_STARTED",

    "expected_answer_generation":
        "NOT_STARTED",

    "citation_adjudication":
        "NOT_STARTED",

    "active_state_mutation":
        "NONE",

    "artifact_write":
        "NONE",
}


cell80_four_component_contract_sha256 = (
    cell80_sha(
        PILOT_QUESTION_DRAFTING_FOUR_COMPONENT_CONTRACT_V1
    )
)


# ============================================================
# 18) Contract-level validation
# ============================================================

cell80_contract_validation_failures = []


if (
    PILOT_QUESTION_AUTHORING_MODE_V1[
        "automatic_acceptance"
    ]
    is not False
):

    cell80_contract_validation_failures.append(
        "AUTOMATIC_ACCEPTANCE_MUST_BE_FALSE"
    )


if (
    PILOT_QUESTION_AUTHORING_MODE_V1[
        "allowed_external_knowledge"
    ]
    is not False
):

    cell80_contract_validation_failures.append(
        "EXTERNAL_KNOWLEDGE_MUST_BE_FORBIDDEN"
    )


if (
    QUESTION_DRAFT_RECORD_SCHEMA_V1[
        "initial_statuses"
    ][
        "expected_answer_status"
    ]
    !=
    "BLOCKED_PENDING_QUESTION_ACCEPTANCE"
):

    cell80_contract_validation_failures.append(
        "EXPECTED_ANSWER_NOT_BLOCKED_PENDING_ACCEPTANCE"
    )


if (
    QUESTION_DRAFT_RECORD_SCHEMA_V1[
        "initial_statuses"
    ][
        "citation_target_validation_status"
    ]
    !=
    "BLOCKED_PENDING_QUESTION_ACCEPTANCE"
):

    cell80_contract_validation_failures.append(
        "CITATION_VALIDATION_NOT_BLOCKED_PENDING_ACCEPTANCE"
    )


if (
    QUESTION_DRAFT_ACCEPTANCE_RUBRIC_V1[
        "human_review_policy"
    ][
        "required_before_acceptance"
    ]
    is not True
):

    cell80_contract_validation_failures.append(
        "HUMAN_REVIEW_NOT_REQUIRED"
    )


if (
    len(
        cell80_required_dimension_counts
    )
    ==
    0
):

    cell80_contract_validation_failures.append(
        "REQUIRED_DIMENSION_UNIVERSE_EMPTY"
    )


# ============================================================
# 19) ACTIVE state snapshot AFTER
# ============================================================

cell80_active_snapshot_after = (
    cell80_sha(
        {
            variable_name:
                globals()[
                    variable_name
                ]

            for variable_name
            in cell80_active_state_object_names

            if (
                variable_name
                in
                globals()
            )
        }
    )
)


cell80_pilot_runtime_snapshot_after = (
    cell80_sha(
        {
            "question_drafting_pilot_packets":
                cell80_packets,

            "question_drafting_pilot_by_id":
                cell80_packet_index,
        }
    )
)


# ============================================================
# 20) Final readiness
# ============================================================

cell80_four_component_contract_ready = (
    len(
        cell80_prerequisite_failures
    )
    ==
    0

    and

    len(
        cell80_guardrail_failures
    )
    ==
    0

    and

    len(
        cell80_id_contract_failures
    )
    ==
    0

    and

    len(
        cell80_rubric_consistency_failures
    )
    ==
    0

    and

    len(
        cell80_source_rule_consistency_failures
    )
    ==
    0

    and

    len(
        cell80_packet_contract_failures
    )
    ==
    0

    and

    len(
        cell80_active_readiness_failures
    )
    ==
    0

    and

    len(
        cell80_contract_validation_failures
    )
    ==
    0
)


cell80_status = (
    "FOUR_PILOT_DRAFTING_COMPONENTS_FROZEN_"
    "READY_FOR_FIRST_QUESTION_DRAFT_STAGING"

    if (
        cell80_four_component_contract_ready
    )

    else
    "FOUR_PILOT_DRAFTING_COMPONENTS_VALIDATION_FAILED"
)


# ============================================================
# 21) Gates
# ============================================================

cell80_gates = {
    "required_runtime_state_is_available":
        (
            len(
                cell80_prerequisite_failures
            )
            ==
            0
        ),

    "cell79_transient_runtime_dependency_has_been_removed":
        True,

    "design_evidence_is_independently_rederived_from_23_current_packets":
        (
            len(
                cell80_required_dimension_counts
            )
            >
            0

            and

            len(
                cell80_task_family_counts
            )
            >
            0

            and

            len(
                cell80_question_archetype_counts
            )
            >
            0
        ),

    "runtime_guardrail_contract_is_uniform":
        (
            len(
                cell80_guardrail_failures
            )
            ==
            0
        ),

    "authoring_mode_is_explicit_new_nb12_design":
        (
            PILOT_QUESTION_AUTHORING_MODE_V1[
                "design_lineage"
            ]
            ==
            "NEWLY_DEFINED_IN_NB12_CELL80"
        ),

    "authoring_mode_requires_human_adjudication":
        (
            PILOT_QUESTION_AUTHORING_MODE_V1[
                "mode"
            ]
            ==
            "MODEL_DRAFT_WITH_HUMAN_ADJUDICATION"

            and

            PILOT_QUESTION_AUTHORING_MODE_V1[
                "automatic_acceptance"
            ]
            is False
        ),

    "question_draft_record_schema_is_defined":
        (
            QUESTION_DRAFT_RECORD_SCHEMA_V1[
                "schema_version"
            ]
            ==
            "question_draft_record_v1"

            and

            "question_text"
            in
            QUESTION_DRAFT_RECORD_SCHEMA_V1[
                "required_fields"
            ]

            and

            "question_text_sha256"
            in
            QUESTION_DRAFT_RECORD_SCHEMA_V1[
                "required_fields"
            ]
        ),

    "question_draft_id_is_deterministic_and_revision_sensitive":
        (
            len(
                cell80_id_contract_failures
            )
            ==
            0
        ),

    "runtime_guardrail_mapping_is_exact":
        (
            isinstance(
                cell80_frozen_guardrails,
                dict,
            )

            and

            set(
                CELL80_RUNTIME_TO_RUBRIC_GUARDRAIL_MAP.keys()
            )
            ==
            set(
                cell80_frozen_guardrails.keys()
            )
        ),

    "runtime_guardrails_are_represented_in_rubric":
        (
            len(
                cell80_rubric_consistency_failures
            )
            ==
            0
        ),

    "source_question_rules_are_represented_in_rubric":
        (
            len(
                cell80_source_rule_consistency_failures
            )
            ==
            0
        ),

    "all_23_current_packets_satisfy_drafting_input_contract":
        (
            len(
                cell80_packet_contract_failures
            )
            ==
            0
        ),

    "all_23_active_tasks_are_ready_with_validated_targets":
        (
            len(
                cell80_active_readiness_failures
            )
            ==
            0
        ),

    "acceptance_requires_human_decision":
        (
            QUESTION_DRAFT_ACCEPTANCE_RUBRIC_V1[
                "human_review_policy"
            ][
                "required_before_acceptance"
            ]
            is True

            and

            QUESTION_DRAFT_ACCEPTANCE_RUBRIC_V1[
                "acceptance_condition"
            ]
            ==
            "ALL_HARD_GATES_PASS_AND_ALL_SEMANTIC_GATES_PASS_"
            "AND_HUMAN_DECISION_ACCEPT"
        ),

    "expected_answer_and_citation_remain_blocked_until_question_acceptance":
        (
            QUESTION_DRAFT_RECORD_SCHEMA_V1[
                "initial_statuses"
            ][
                "expected_answer_status"
            ]
            ==
            "BLOCKED_PENDING_QUESTION_ACCEPTANCE"

            and

            QUESTION_DRAFT_RECORD_SCHEMA_V1[
                "initial_statuses"
            ][
                "citation_target_validation_status"
            ]
            ==
            "BLOCKED_PENDING_QUESTION_ACCEPTANCE"
        ),

    "full_572_generation_remains_blocked":
        (
            PILOT_QUESTION_DRAFTING_FOUR_COMPONENT_CONTRACT_V1[
                "full_population_generation"
            ]
            ==
            "BLOCKED_PENDING_23_PILOT_ACCEPTANCE"
        ),

    "four_component_contract_has_zero_validation_failures":
        (
            len(
                cell80_contract_validation_failures
            )
            ==
            0
        ),

    "four_component_contract_is_frozen":
        (
            cell80_status
            ==
            "FOUR_PILOT_DRAFTING_COMPONENTS_FROZEN_"
            "READY_FOR_FIRST_QUESTION_DRAFT_STAGING"
        ),

    "active_state_remains_byte_stable":
        (
            cell80_active_snapshot_before
            ==
            cell80_active_snapshot_after
        ),

    "pilot_runtime_remains_byte_stable":
        (
            cell80_pilot_runtime_snapshot_before
            ==
            cell80_pilot_runtime_snapshot_after
        ),

    "question_generation_not_started":
        (
            PILOT_QUESTION_DRAFTING_FOUR_COMPONENT_CONTRACT_V1[
                "question_generation"
            ]
            ==
            "NOT_STARTED"
        ),

    "expected_answer_generation_not_started":
        (
            PILOT_QUESTION_DRAFTING_FOUR_COMPONENT_CONTRACT_V1[
                "expected_answer_generation"
            ]
            ==
            "NOT_STARTED"
        ),

    "citation_adjudication_not_started":
        (
            PILOT_QUESTION_DRAFTING_FOUR_COMPONENT_CONTRACT_V1[
                "citation_adjudication"
            ]
            ==
            "NOT_STARTED"
        ),

    "no_artifact_write":
        (
            PILOT_QUESTION_DRAFTING_FOUR_COMPONENT_CONTRACT_V1[
                "artifact_write"
            ]
            ==
            "NONE"
        ),
}


cell_80_pass = all(
    cell80_gates.values()
)


# ============================================================
# 22) Output
# ============================================================

print(
    "=== PILOT QUESTION DRAFTING — FOUR COMPONENT DESIGN FREEZE — REPAIRED V2 ==="
)


print(
    "\n=== RUNTIME / LINEAGE ==="
)


print(
    "Missing runtime objects:",
    cell80_missing_runtime_objects
)


print(
    "Cell79 runtime lineage:",
    cell80_cell79_lineage_status
)


print(
    "Cell79 contract status:",
    cell80_cell79_contract_status
)


print(
    "Prerequisite failures:",
    cell80_prerequisite_failures
)


print(
    "\n=== INDEPENDENTLY REDERIVED PILOT EVIDENCE ==="
)


print(
    "Pilot packets:",
    (
        len(
            cell80_packets
        )
        if isinstance(
            cell80_packets,
            list,
        )
        else None
    )
)


print(
    "Required dimensions:",
    cell80_required_dimension_counts
)


print(
    "Task families:",
    cell80_task_family_counts
)


print(
    "Question archetypes:",
    cell80_question_archetype_counts
)


print(
    "Evidence scopes:",
    cell80_evidence_scope_counts
)


print(
    "\n=== RUNTIME GUARDRAILS ==="
)


print(
    cell80_frozen_guardrails
)


print(
    "Failures:",
    cell80_guardrail_failures
)


print(
    "\n=== 1) AUTHORING MODE ==="
)


print(
    PILOT_QUESTION_AUTHORING_MODE_V1
)


print(
    "\n=== 2) QUESTION DRAFT SCHEMA ==="
)


print(
    "Schema version:",
    QUESTION_DRAFT_RECORD_SCHEMA_V1[
        "schema_version"
    ]
)


print(
    "Initial statuses:",
    QUESTION_DRAFT_RECORD_SCHEMA_V1[
        "initial_statuses"
    ]
)


print(
    "\n=== 3) QUESTION DRAFT ID ==="
)


print(
    "Test A:",
    cell80_id_test_a
)


print(
    "Test B:",
    cell80_id_test_b
)


print(
    "Revision:",
    cell80_id_test_revision
)


print(
    "Failures:",
    cell80_id_contract_failures
)


print(
    "\n=== 4) ACCEPTANCE RUBRIC ==="
)


print(
    "Runtime → Rubric mapping:"
)


for runtime_key, rubric_key in (
    CELL80_RUNTIME_TO_RUBRIC_GUARDRAIL_MAP.items()
):

    print(
        "-",
        runtime_key,
        "→",
        rubric_key
    )


print(
    "Runtime mapping failures:",
    cell80_rubric_consistency_failures
)


print(
    "Source-rule failures:",
    cell80_source_rule_consistency_failures
)


print(
    "Dimension universe:",
    QUESTION_DRAFT_ACCEPTANCE_RUBRIC_V1[
        "dimension_policy"
    ][
        "global_dimension_universe"
    ]
)


print(
    "\n=== ACTIVE 23-PILOT READINESS ==="
)


print(
    "Resolved tasks:",
    len(
        cell80_resolved_pilot_tasks
    )
)


print(
    "Resolved targets:",
    len(
        cell80_resolved_pilot_targets
    )
)


print(
    "Task construction:",
    cell80_task_construction_counts
)


print(
    "Target status:",
    cell80_task_target_status_counts
)


print(
    "Readiness failures:",
    cell80_active_readiness_failures
)


print(
    "\n=== PACKET CONTRACT ==="
)


print(
    "Failures:",
    cell80_packet_contract_failures
)


print(
    "\n=== CONTRACT VALIDATION ==="
)


print(
    "Failures:",
    cell80_contract_validation_failures
)


print(
    "Contract SHA256:",
    cell80_four_component_contract_sha256
)


print(
    "STATUS:",
    cell80_status
)


print(
    "\n=== CODE CELL 80 GATES ==="
)


for gate_name, passed in (
    cell80_gates.items()
):

    print(
        gate_name,
        ":",
        "PASS"
        if passed
        else "FAIL"
    )


print(
    "\nFAILED GATES:",
    [
        gate_name

        for gate_name, passed
        in cell80_gates.items()

        if not passed
    ]
)


print(
    "\nCODE CELL 80 GATE:",
    "PASS"
    if cell_80_pass
    else "FAIL"
)


assert cell_80_pass, (
    "The four Pilot Question-Drafting components are not "
    "safely frozen. Do not generate a question. Repair this "
    "SAME Code Cell 80 using the printed runtime, guardrail, "
    "packet, readiness, ID and rubric diagnostics."
)
```

<li>هیچ Question یا Answer تولید نشده باشد؛</li>
<li>Active State و Pilot Runtime قبل و بعد Byte-Stable باقی بمانند.</li>
</ul>

<p>
خروجی این مرحله یک <code>First Question Authoring Packet</code> کامل و Evidence-Bounded خواهد بود که در مرحله‌ی بعد مبنای ساخت اولین Question Draft واقعی قرار می‌گیرد.
</p>

</div>

```python
# Code Cell 81
#
# ============================================================
# First Pilot Question Authoring Packet Staging
#
# ONE deterministic Pilot only.
#
# NO question generation.
# NO answer generation.
# NO citation adjudication.
# NO active-state mutation.
# NO artifact write.
# ============================================================


# ============================================================
# 1) Helpers
# ============================================================

def cell81_sha(value):

    return hashlib.sha256(
        stable_json_key(
            value
        ).encode(
            "utf-8"
        )
    ).hexdigest()


# ============================================================
# 2) Required upstream contract
# ============================================================

cell81_prerequisite_failures = []


if (
    globals().get(
        "cell_80_pass"
    )
    is not True
):

    cell81_prerequisite_failures.append(
        "CELL80_NOT_PASS"
    )


if (
    globals().get(
        "cell80_status"
    )
    !=
    (
        "FOUR_PILOT_DRAFTING_COMPONENTS_FROZEN_"
        "READY_FOR_FIRST_QUESTION_DRAFT_STAGING"
    )
):

    cell81_prerequisite_failures.append(
        {
            "reason":
                "CELL80_STATUS_UNEXPECTED",

            "actual":
                globals().get(
                    "cell80_status"
                ),
        }
    )


CELL81_REQUIRED_RUNTIME_OBJECTS = [
    "question_drafting_pilot_packets",
    "question_drafting_pilot_by_id",
    "task_draft_plan_rows",
    "target_evidence_records",
    "PILOT_QUESTION_AUTHORING_MODE_V1",
    "QUESTION_DRAFT_RECORD_SCHEMA_V1",
    "QUESTION_DRAFT_ID_CONTRACT_V1",
    "QUESTION_DRAFT_ACCEPTANCE_RUBRIC_V1",
    "PILOT_QUESTION_DRAFTING_FOUR_COMPONENT_CONTRACT_V1",
]


cell81_missing_runtime_objects = [
    variable_name

    for variable_name
    in CELL81_REQUIRED_RUNTIME_OBJECTS

    if (
        variable_name
        not in
        globals()
    )
]


if (
    cell81_missing_runtime_objects
):

    cell81_prerequisite_failures.append(
        {
            "reason":
                "REQUIRED_RUNTIME_OBJECTS_MISSING",

            "objects":
                cell81_missing_runtime_objects,
        }
    )


# ============================================================
# 3) Snapshot active state BEFORE
# ============================================================

CELL81_ACTIVE_STATE_OBJECT_NAMES = [
    "golden_case_records",
    "case_groundability_precheck",
    "case_task_design",
    "task_draft_plan_rows",
    "target_evidence_records",
    "pilot_selection_rows",
    "golden_case_records_by_id",
    "task_draft_plans",
    "target_evidence_by_case_id",
    "target_evidence_by_task_draft_id",
]


cell81_active_snapshot_before = (
    cell81_sha(
        {
            variable_name:
                globals()[
                    variable_name
                ]

            for variable_name
            in CELL81_ACTIVE_STATE_OBJECT_NAMES

            if (
                variable_name
                in
                globals()
            )
        }
    )
)


cell81_pilot_snapshot_before = (
    cell81_sha(
        {
            "question_drafting_pilot_packets":
                question_drafting_pilot_packets,

            "question_drafting_pilot_by_id":
                question_drafting_pilot_by_id,
        }
    )
)


# ============================================================
# 4) Deterministic first-pilot selection rule
#
# Start with the simplest direct Primary Website retrieval case.
# ============================================================

CELL81_FIRST_PILOT_SELECTION_RULE = {
    "source_type":
        "website",

    "stratum":
        "direct_author_short",

    "task_family":
        "DIRECT_PRIMARY_SOURCE_RETRIEVAL",

    "question_archetype":
        "AUTHORITY_CONSTRAINED_RETRIEVAL",
}


cell81_first_pilot_candidates = [
    packet

    for packet
    in question_drafting_pilot_packets

    if (
        packet.get(
            "source_type"
        )
        ==
        CELL81_FIRST_PILOT_SELECTION_RULE[
            "source_type"
        ]

        and

        packet.get(
            "stratum"
        )
        ==
        CELL81_FIRST_PILOT_SELECTION_RULE[
            "stratum"
        ]

        and

        packet.get(
            "task_family"
        )
        ==
        CELL81_FIRST_PILOT_SELECTION_RULE[
            "task_family"
        ]

        and

        packet.get(
            "question_archetype"
        )
        ==
        CELL81_FIRST_PILOT_SELECTION_RULE[
            "question_archetype"
        ]
    )
]


cell81_selection_failures = []


if (
    len(
        cell81_first_pilot_candidates
    )
    !=
    1
):

    cell81_selection_failures.append(
        {
            "reason":
                "FIRST_PILOT_SELECTION_NOT_UNIQUE",

            "expected":
                1,

            "actual":
                len(
                    cell81_first_pilot_candidates
                ),
        }
    )


cell81_selected_packet = (
    cell81_first_pilot_candidates[
        0
    ]

    if (
        len(
            cell81_first_pilot_candidates
        )
        ==
        1
    )

    else None
)


# ============================================================
# 5) Resolve corresponding Active Task and Target
# ============================================================

cell81_selected_task_rows = []

cell81_selected_target_rows = []


if (
    cell81_selected_packet
    is not None
):

    cell81_selected_task_rows = [
        row

        for row
        in task_draft_plan_rows

        if (
            row.get(
                "task_draft_id"
            )
            ==
            cell81_selected_packet.get(
                "task_draft_id"
            )
        )
    ]


    cell81_selected_target_rows = [
        row

        for row
        in target_evidence_records

        if (
            row.get(
                "target_evidence_id"
            )
            ==
            cell81_selected_packet.get(
                "target_evidence_id"
            )
        )
    ]


if (
    len(
        cell81_selected_task_rows
    )
    !=
    1
):

    cell81_selection_failures.append(
        {
            "reason":
                "SELECTED_TASK_NOT_UNIQUE",

            "actual":
                len(
                    cell81_selected_task_rows
                ),
        }
    )


if (
    len(
        cell81_selected_target_rows
    )
    !=
    1
):

    cell81_selection_failures.append(
        {
            "reason":
                "SELECTED_TARGET_NOT_UNIQUE",

            "actual":
                len(
                    cell81_selected_target_rows
                ),
        }
    )


cell81_selected_task = (
    cell81_selected_task_rows[
        0
    ]

    if (
        len(
            cell81_selected_task_rows
        )
        ==
        1
    )

    else None
)


cell81_selected_target = (
    cell81_selected_target_rows[
        0
    ]

    if (
        len(
            cell81_selected_target_rows
        )
        ==
        1
    )

    else None
)


# ============================================================
# 6) Validate Task / Target lifecycle
# ============================================================

cell81_readiness_failures = []


if (
    cell81_selected_task
    is not None
):

    if (
        cell81_selected_task.get(
            "construction_eligibility"
        )
        !=
        "READY_FOR_QUESTION_DRAFT"
    ):

        cell81_readiness_failures.append(
            {
                "reason":
                    "TASK_NOT_READY_FOR_QUESTION_DRAFT",

                "actual":
                    cell81_selected_task.get(
                        "construction_eligibility"
                    ),
            }
        )


    if (
        cell81_selected_task.get(
            "target_evidence_status"
        )
        !=
        "SELECTED_AND_VALIDATED"
    ):

        cell81_readiness_failures.append(
            {
                "reason":
                    "TASK_TARGET_NOT_SELECTED_AND_VALIDATED",

                "actual":
                    cell81_selected_task.get(
                        "target_evidence_status"
                    ),
            }
        )


# ============================================================
# 7) Independently validate staged Evidence text
# ============================================================

cell81_evidence_failures = []

cell81_target_text_parts = (
    cell81_selected_packet.get(
        "target_text_parts",
        [],
    )

    if (
        cell81_selected_packet
        is not None
    )

    else []
)


if not (
    isinstance(
        cell81_target_text_parts,
        list,
    )

    and

    len(
        cell81_target_text_parts
    )
    >
    0
):

    cell81_evidence_failures.append(
        "TARGET_TEXT_PARTS_EMPTY"
    )


for part_index, part in enumerate(
    cell81_target_text_parts
):

    target_text = (
        part.get(
            "text"
        )
    )


    if not (
        isinstance(
            target_text,
            str,
        )

        and

        target_text.strip()
    ):

        cell81_evidence_failures.append(
            {
                "reason":
                    "TARGET_TEXT_EMPTY",

                "part_index":
                    part_index,
            }
        )

        continue


    recomputed_sha = (
        hashlib.sha256(
            target_text.encode(
                "utf-8"
            )
        ).hexdigest()
    )


    if (
        recomputed_sha
        !=
        part.get(
            "text_sha256"
        )
    ):

        cell81_evidence_failures.append(
            {
                "reason":
                    "TARGET_TEXT_SHA_MISMATCH",

                "part_index":
                    part_index,

                "expected":
                    part.get(
                        "text_sha256"
                    ),

                "actual":
                    recomputed_sha,
            }
        )


# ============================================================
# 8) Validate frozen Guardrails
# ============================================================

CELL81_EXPECTED_GUARDRAILS = {
    "external_knowledge":
        "FORBIDDEN",

    "unknown_speaker_inference":
        "FORBIDDEN",

    "unsupported_attribution":
        "FORBIDDEN",

    "secondary_to_primary_promotion":
        "FORBIDDEN",

    "locator_fabrication":
        "FORBIDDEN",

    "answer_leakage_in_question":
        "FORBIDDEN",

    "internal_ids_in_question":
        "FORBIDDEN",
}


cell81_guardrail_failures = []


if (
    cell81_selected_packet
    is not None

    and

    cell81_selected_packet.get(
        "guardrails"
    )
    !=
    CELL81_EXPECTED_GUARDRAILS
):

    cell81_guardrail_failures.append(
        {
            "reason":
                "SELECTED_PACKET_GUARDRAILS_UNEXPECTED",

            "actual":
                cell81_selected_packet.get(
                    "guardrails"
                ),
        }
    )


# ============================================================
# 9) Verify pre-generation state
# ============================================================

CELL81_EXPECTED_PREGEN_STATUS = {
    "question_draft_status":
        "NOT_STARTED",

    "question_groundability_status":
        "NOT_STARTED",

    "expected_answer_status":
        "NOT_STARTED",

    "citation_target_validation_status":
        "NOT_STARTED",
}


cell81_pregen_failures = []


if (
    cell81_selected_packet
    is not None
):

    for field_name, expected_value in (
        CELL81_EXPECTED_PREGEN_STATUS.items()
    ):

        actual_value = (
            cell81_selected_packet.get(
                field_name
            )
        )


        if (
            actual_value
            !=
            expected_value
        ):

            cell81_pregen_failures.append(
                {
                    "reason":
                        "PREGEN_STATUS_MISMATCH",

                    "field":
                        field_name,

                    "expected":
                        expected_value,

                    "actual":
                        actual_value,
                }
            )


# ============================================================
# 10) Build read-only Authoring Packet
#
# This is NOT a Question Draft Record.
# No question_text is created here.
# ============================================================

cell81_first_question_authoring_packet = (
    {
        "staging_version":
            "first_question_authoring_packet_v1",

        "selection_rule":
            dict(
                CELL81_FIRST_PILOT_SELECTION_RULE
            ),

        "pilot_case_id":
            cell81_selected_packet[
                "pilot_case_id"
            ],

        "golden_case_id":
            cell81_selected_packet[
                "golden_case_id"
            ],

        "task_draft_id":
            cell81_selected_packet[
                "task_draft_id"
            ],

        "target_evidence_id":
            cell81_selected_packet[
                "target_evidence_id"
            ],

        "source_type":
            cell81_selected_packet[
                "source_type"
            ],

        "stratum":
            cell81_selected_packet[
                "stratum"
            ],

        "task_family":
            cell81_selected_packet[
                "task_family"
            ],

        "question_archetype":
            cell81_selected_packet[
                "question_archetype"
            ],

        "required_dimensions":
            list(
                cell81_selected_packet[
                    "required_dimensions"
                ]
            ),

        "authority_expectation":
            cell81_selected_packet[
                "authority_expectation"
            ],

        "context_expectation":
            cell81_selected_packet[
                "context_expectation"
            ],

        "evidence_scope":
            cell81_selected_packet[
                "evidence_scope"
            ],

        "target_selection_method":
            cell81_selected_packet[
                "target_selection_method"
            ],

        "target_selection_basis":
            cell81_selected_packet[
                "target_selection_basis"
            ],

        "target_total_tokens":
            cell81_selected_packet[
                "target_total_tokens"
            ],

        "target_text_parts":
            [
                {
                    "part_index":
                        part_index,

                    "role":
                        part[
                            "role"
                        ],

                    "text":
                        part[
                            "text"
                        ],

                    "text_sha256":
                        part[
                            "text_sha256"
                        ],

                    "locator":
                        part[
                            "locator"
                        ],

                    "attribution":
                        part[
                            "attribution"
                        ],

                    "speaker_state":
                        part[
                            "speaker_state"
                        ],
                }

                for part_index, part
                in enumerate(
                    cell81_target_text_parts
                )
            ],

        "guardrails":
            dict(
                cell81_selected_packet[
                    "guardrails"
                ]
            ),

        "authoring_mode":
            PILOT_QUESTION_AUTHORING_MODE_V1[
                "mode"
            ],

        "allowed_evidence_source":
            PILOT_QUESTION_AUTHORING_MODE_V1[
                "allowed_evidence_source"
            ],

        "question_text":
            None,

        "question_generation_status":
            "NOT_STARTED",

        "expected_answer_generation":
            "BLOCKED",

        "citation_adjudication":
            "BLOCKED",

        "full_572_generation":
            "BLOCKED",
    }

    if (
        cell81_selected_packet
        is not None
    )

    else None
)


cell81_authoring_packet_sha256 = (
    cell81_sha(
        cell81_first_question_authoring_packet
    )

    if (
        cell81_first_question_authoring_packet
        is not None
    )

    else None
)


# ============================================================
# 11) Snapshot AFTER — prove no active mutation
# ============================================================

cell81_active_snapshot_after = (
    cell81_sha(
        {
            variable_name:
                globals()[
                    variable_name
                ]

            for variable_name
            in CELL81_ACTIVE_STATE_OBJECT_NAMES

            if (
                variable_name
                in
                globals()
            )
        }
    )
)


cell81_pilot_snapshot_after = (
    cell81_sha(
        {
            "question_drafting_pilot_packets":
                question_drafting_pilot_packets,

            "question_drafting_pilot_by_id":
                question_drafting_pilot_by_id,
        }
    )
)


# ============================================================
# 12) Final status and gates
# ============================================================

cell81_status = (
    "FIRST_PILOT_AUTHORING_PACKET_STAGED_NO_QUESTION_GENERATION"

    if (
        len(
            cell81_prerequisite_failures
        )
        ==
        0

        and

        len(
            cell81_selection_failures
        )
        ==
        0

        and

        len(
            cell81_readiness_failures
        )
        ==
        0

        and

        len(
            cell81_evidence_failures
        )
        ==
        0

        and

        len(
            cell81_guardrail_failures
        )
        ==
        0

        and

        len(
            cell81_pregen_failures
        )
        ==
        0
    )

    else
    "FIRST_PILOT_AUTHORING_PACKET_STAGING_FAILED"
)


cell81_gates = {
    "cell80_is_pass":
        (
            globals().get(
                "cell_80_pass"
            )
            is True
        ),

    "cell80_is_ready_for_first_question_staging":
        (
            globals().get(
                "cell80_status"
            )
            ==
            (
                "FOUR_PILOT_DRAFTING_COMPONENTS_FROZEN_"
                "READY_FOR_FIRST_QUESTION_DRAFT_STAGING"
            )
        ),

    "exactly_one_first_pilot_is_selected":
        (
            len(
                cell81_first_pilot_candidates
            )
            ==
            1
        ),

    "selected_task_is_ready":
        (
            len(
                cell81_readiness_failures
            )
            ==
            0
        ),

    "selected_target_text_is_nonempty_and_hash_valid":
        (
            len(
                cell81_evidence_failures
            )
            ==
            0
        ),

    "frozen_guardrails_are_preserved":
        (
            len(
                cell81_guardrail_failures
            )
            ==
            0
        ),

    "selected_packet_is_still_pregeneration":
        (
            len(
                cell81_pregen_failures
            )
            ==
            0
        ),

    "question_text_has_not_been_generated":
        (
            cell81_first_question_authoring_packet
            is not None

            and

            cell81_first_question_authoring_packet[
                "question_text"
            ]
            is None

            and

            cell81_first_question_authoring_packet[
                "question_generation_status"
            ]
            ==
            "NOT_STARTED"
        ),

    "expected_answer_remains_blocked":
        (
            cell81_first_question_authoring_packet
            is not None

            and

            cell81_first_question_authoring_packet[
                "expected_answer_generation"
            ]
            ==
            "BLOCKED"
        ),

    "citation_adjudication_remains_blocked":
        (
            cell81_first_question_authoring_packet
            is not None

            and

            cell81_first_question_authoring_packet[
                "citation_adjudication"
            ]
            ==
            "BLOCKED"
        ),

    "full_572_generation_remains_blocked":
        (
            cell81_first_question_authoring_packet
            is not None

            and

            cell81_first_question_authoring_packet[
                "full_572_generation"
            ]
            ==
            "BLOCKED"
        ),

    "active_state_remains_byte_stable":
        (
            cell81_active_snapshot_before
            ==
            cell81_active_snapshot_after
        ),

    "pilot_runtime_remains_byte_stable":
        (
            cell81_pilot_snapshot_before
            ==
            cell81_pilot_snapshot_after
        ),

    "no_artifact_write":
        True,
}


cell_81_pass = all(
    cell81_gates.values()
)


# ============================================================
# 13) Diagnostics
# ============================================================

print(
    "=== FIRST PILOT QUESTION AUTHORING PACKET STAGING ==="
)


print(
    "\n=== PREREQUISITES ==="
)

print(
    "Failures:",
    cell81_prerequisite_failures
)


print(
    "\n=== SELECTION RULE ==="
)

print(
    CELL81_FIRST_PILOT_SELECTION_RULE
)

print(
    "Candidates:",
    len(
        cell81_first_pilot_candidates
    )
)


print(
    "\n=== SELECTED PILOT ==="
)

if (
    cell81_first_question_authoring_packet
    is not None
):

    for field_name in [
        "pilot_case_id",
        "golden_case_id",
        "task_draft_id",
        "target_evidence_id",
        "source_type",
        "stratum",
        "task_family",
        "question_archetype",
        "required_dimensions",
        "authority_expectation",
        "context_expectation",
        "evidence_scope",
        "target_selection_method",
        "target_selection_basis",
        "target_total_tokens",
        "authoring_mode",
    ]:

        print(
            field_name,
            ":",
            cell81_first_question_authoring_packet[
                field_name
            ]
        )


print(
    "\n=== TARGET EVIDENCE TEXT ==="
)


for part in (
    (
        cell81_first_question_authoring_packet[
            "target_text_parts"
        ]

        if (
            cell81_first_question_authoring_packet
            is not None
        )

        else []
    )
):

    print(
        "\n--- PART",
        part[
            "part_index"
        ],
        "---"
    )

    print(
        "Role:",
        part[
            "role"
        ]
    )

    print(
        "Text:"
    )

    print(
        part[
            "text"
        ]
    )

    print(
        "Attribution:",
        part[
            "attribution"
        ]
    )

    print(
        "Speaker state:",
        part[
            "speaker_state"
        ]
    )


print(
    "\n=== GUARDRAILS ==="
)

print(
    (
        cell81_first_question_authoring_packet[
            "guardrails"
        ]

        if (
            cell81_first_question_authoring_packet
            is not None
        )

        else None
    )
)


print(
    "\n=== STAGING ==="
)

print(
    "Authoring packet SHA256:",
    cell81_authoring_packet_sha256
)

print(
    "Question generation:",
    (
        cell81_first_question_authoring_packet[
            "question_generation_status"
        ]

        if (
            cell81_first_question_authoring_packet
            is not None
        )

        else None
    )
)

print(
    "Expected answer:",
    (
        cell81_first_question_authoring_packet[
            "expected_answer_generation"
        ]

        if (
            cell81_first_question_authoring_packet
            is not None
        )

        else None
    )
)

print(
    "Citation adjudication:",
    (
        cell81_first_question_authoring_packet[
            "citation_adjudication"
        ]

        if (
            cell81_first_question_authoring_packet
            is not None
        )

        else None
    )
)


print(
    "\nSelection failures:",
    cell81_selection_failures
)

print(
    "Readiness failures:",
    cell81_readiness_failures
)

print(
    "Evidence failures:",
    cell81_evidence_failures
)

print(
    "Guardrail failures:",
    cell81_guardrail_failures
)

print(
    "Pregeneration failures:",
    cell81_pregen_failures
)


print(
    "\nSTATUS:",
    cell81_status
)


print(
    "\n=== CODE CELL 81 GATES ==="
)


for gate_name, gate_value in (
    cell81_gates.items()
):

    print(
        gate_name,
        ":",
        (
            "PASS"
            if gate_value
            else "FAIL"
        )
    )


print(
    "\nFAILED GATES:",
    [
        gate_name

        for gate_name, gate_value
        in cell81_gates.items()

        if not gate_value
    ]
)


print(
    "\nCODE CELL 81 GATE:",
    (
        "PASS"
        if cell_81_pass
        else "FAIL"
    )
)


assert cell_81_pass, (
    "First Pilot Question Authoring Packet staging failed. "
    "Do not generate a question. Repair this SAME Code Cell 81 "
    "using the printed diagnostics."
)
```

```python
# Code Cell 82

# ============================================================
# First real Question Draft materialization
#
# No human acceptance.
# No expected-answer generation.
# No citation adjudication.
# No artifact write.
# ============================================================


# ============================================================
# 1) Prerequisites
# ============================================================

cell82_prerequisite_failures = []


if (
    globals().get(
        "cell_81_pass"
    )
    is not True
):

    cell82_prerequisite_failures.append(
        "CELL81_NOT_PASS"
    )


if (
    globals().get(
        "cell81_status"
    )
    !=
    "FIRST_PILOT_AUTHORING_PACKET_STAGED_NO_QUESTION_GENERATION"
):

    cell82_prerequisite_failures.append(
        {
            "reason":
                "CELL81_STATUS_UNEXPECTED",

            "actual":
                globals().get(
                    "cell81_status"
                ),
        }
    )


CELL82_REQUIRED_RUNTIME_OBJECTS = [
    "cell81_first_question_authoring_packet",
    "cell81_authoring_packet_sha256",
    "QUESTION_DRAFT_RECORD_SCHEMA_V1",
    "QUESTION_DRAFT_ID_CONTRACT_V1",
    "QUESTION_DRAFT_ACCEPTANCE_RUBRIC_V1",
    "PILOT_QUESTION_AUTHORING_MODE_V1",
]


cell82_missing_runtime_objects = [
    name

    for name
    in CELL82_REQUIRED_RUNTIME_OBJECTS

    if (
        name
        not in
        globals()
    )
]


if (
    cell82_missing_runtime_objects
):

    cell82_prerequisite_failures.append(
        {
            "reason":
                "REQUIRED_RUNTIME_OBJECTS_MISSING",

            "objects":
                cell82_missing_runtime_objects,
        }
    )


# ============================================================
# 2) Freeze active state before drafting
# ============================================================

CELL82_ACTIVE_STATE_OBJECT_NAMES = [
    "golden_case_records",
    "case_groundability_precheck",
    "case_task_design",
    "task_draft_plan_rows",
    "target_evidence_records",
    "pilot_selection_rows",
    "golden_case_records_by_id",
    "task_draft_plans",
    "target_evidence_by_case_id",
    "target_evidence_by_task_draft_id",
]


def cell82_sha(value):

    return hashlib.sha256(
        stable_json_key(
            value
        ).encode(
            "utf-8"
        )
    ).hexdigest()


cell82_active_snapshot_before = (
    cell82_sha(
        {
            name:
                globals()[
                    name
                ]

            for name
            in CELL82_ACTIVE_STATE_OBJECT_NAMES

            if (
                name
                in
                globals()
            )
        }
    )
)


cell82_pilot_snapshot_before = (
    cell82_sha(
        {
            "question_drafting_pilot_packets":
                question_drafting_pilot_packets,

            "question_drafting_pilot_by_id":
                question_drafting_pilot_by_id,
        }
    )
)


# ============================================================
# 3) Frozen first-pilot input
# ============================================================

cell82_packet = (
    cell81_first_question_authoring_packet
)


cell82_packet_failures = []


CELL82_EXPECTED_SELECTION = {
    "source_type":
        "website",

    "stratum":
        "direct_author_short",

    "task_family":
        "DIRECT_PRIMARY_SOURCE_RETRIEVAL",

    "question_archetype":
        "AUTHORITY_CONSTRAINED_RETRIEVAL",

    "authority_expectation":
        "PRIMARY_DIRECT_AUTHOR",

    "context_expectation":
        "MINIMAL_DIRECT_EVIDENCE",

    "evidence_scope":
        "SINGLE_SOURCE_DOCUMENT",
}


for field_name, expected_value in (
    CELL82_EXPECTED_SELECTION.items()
):

    actual_value = (
        cell82_packet.get(
            field_name
        )
    )


    if (
        actual_value
        !=
        expected_value
    ):

        cell82_packet_failures.append(
            {
                "reason":
                    "FIRST_PILOT_CONTRACT_MISMATCH",

                "field":
                    field_name,

                "expected":
                    expected_value,

                "actual":
                    actual_value,
            }
        )


# ============================================================
# 4) First authored Question
#
# Evidence-bounded model draft.
# ============================================================

CELL82_QUESTION_TEXT = (
    "بر اساس توضیح دکتر کمیل رودی، برای شروع معاملات در بورس "
    "چه مراحلی باید طی شود و پس از آن چه روش‌هایی برای خرید و فروش "
    "سهام ذکر شده است؟"
)


cell82_question_text_sha256 = (
    hashlib.sha256(
        CELL82_QUESTION_TEXT.encode(
            "utf-8"
        )
    ).hexdigest()
)


# ============================================================
# 5) Evidence-bound authoring preflight
# ============================================================

cell82_question_preflight = {}


cell82_question_preflight[
    "natural_nonempty_question"
] = (
    isinstance(
        CELL82_QUESTION_TEXT,
        str,
    )

    and

    len(
        CELL82_QUESTION_TEXT.strip()
    )
    >=
    20

    and

    CELL82_QUESTION_TEXT.endswith(
        "؟"
    )
)


cell82_internal_markers = [
    "golden-case:",
    "task-draft:",
    "target-evidence:",
    "question-pilot:",
    "website:",
    "program:",
    "transcript-file:",
    "book:",
    "sha256",
]


cell82_question_preflight[
    "no_internal_ids_or_hash_markers"
] = not any(
    marker
    in
    CELL82_QUESTION_TEXT

    for marker
    in cell82_internal_markers
)


cell82_question_preflight[
    "no_url_or_locator_fabrication"
] = (
    "http://"
    not in
    CELL82_QUESTION_TEXT

    and

    "https://"
    not in
    CELL82_QUESTION_TEXT

    and

    "www."
    not in
    CELL82_QUESTION_TEXT
)


cell82_target_parts = (
    cell82_packet.get(
        "target_text_parts",
        [],
    )
)


cell82_joined_evidence_text = (
    "\n".join(
        part[
            "text"
        ]

        for part
        in cell82_target_parts
    )
)


cell82_attribution_names = set()


for part in (
    cell82_target_parts
):

    attribution = (
        part.get(
            "attribution"
        )
        or
        {}
    )


    for attribution_field in [
        "author",
        "credited_author_name",
        "knowledge_owner",
    ]:

        attribution_value = (
            attribution.get(
                attribution_field
            )
        )


        if (
            isinstance(
                attribution_value,
                str,
            )

            and

            attribution_value.strip()
        ):

            cell82_attribution_names.add(
                attribution_value.strip()
            )


cell82_question_preflight[
    "named_author_is_supported_by_attribution"
] = (
    "دکتر کمیل رودی"
    in
    cell82_attribution_names
)


cell82_direct_author_parts = [
    part

    for part
    in cell82_target_parts

    if (
        (
            part.get(
                "attribution"
            )
            or
            {}
        ).get(
            "content_authority_type"
        )
        ==
        "direct_author"

        and

        (
            part.get(
                "attribution"
            )
            or
            {}
        ).get(
            "attribution_status"
        )
        ==
        "policy_validated"
    )
]


cell82_question_preflight[
    "primary_direct_author_authority_is_supported"
] = (
    len(
        cell82_direct_author_parts
    )
    >=
    1
)


cell82_required_evidence_anchors = [
    "بورس",
    "معاملات",
    "خرید و فروش سهام",
]


cell82_question_preflight[
    "question_topic_is_present_in_target_evidence"
] = all(
    anchor
    in
    cell82_joined_evidence_text

    for anchor
    in cell82_required_evidence_anchors
)


cell82_question_preflight[
    "question_does_not_copy_full_evidence"
] = (
    CELL82_QUESTION_TEXT
    not in
    cell82_joined_evidence_text

    and

    cell82_joined_evidence_text
    not in
    CELL82_QUESTION_TEXT
)


cell82_question_preflight[
    "frozen_guardrails_are_present"
] = (
    cell82_packet.get(
        "guardrails"
    )
    ==
    {
        "external_knowledge":
            "FORBIDDEN",

        "unknown_speaker_inference":
            "FORBIDDEN",

        "unsupported_attribution":
            "FORBIDDEN",

        "secondary_to_primary_promotion":
            "FORBIDDEN",

        "locator_fabrication":
            "FORBIDDEN",

        "answer_leakage_in_question":
            "FORBIDDEN",

        "internal_ids_in_question":
            "FORBIDDEN",
    }
)


cell82_preflight_failures = [
    gate_name

    for gate_name, gate_value
    in cell82_question_preflight.items()

    if not gate_value
]


# ============================================================
# 6) Resolve frozen Question Draft schema
# ============================================================

cell82_schema_version = (
    QUESTION_DRAFT_RECORD_SCHEMA_V1.get(
        "schema_version"
    )

    or

    QUESTION_DRAFT_RECORD_SCHEMA_V1.get(
        "version"
    )
)


cell82_schema_required_fields = (
    QUESTION_DRAFT_RECORD_SCHEMA_V1.get(
        "required_fields",
        [],
    )
)


cell82_schema_failures = []


if (
    cell82_schema_version
    !=
    "question_draft_record_v1"
):

    cell82_schema_failures.append(
        {
            "reason":
                "QUESTION_SCHEMA_VERSION_UNEXPECTED",

            "actual":
                cell82_schema_version,
        }
    )


if not (
    isinstance(
        cell82_schema_required_fields,
        list,
    )

    and

    len(
        cell82_schema_required_fields
    )
    >
    0
):

    cell82_schema_failures.append(
        {
            "reason":
                "QUESTION_SCHEMA_REQUIRED_FIELDS_UNAVAILABLE",

            "actual":
                cell82_schema_required_fields,
        }
    )


# ============================================================
# 7) Discover the frozen deterministic ID builder
# ============================================================

cell82_question_id_builder_candidates = []


for name, value in list(
    globals().items()
):

    if not (
        callable(
            value
        )
    ):

        continue


    lowered_name = (
        name.lower()
    )


    if not (
        "question"
        in
        lowered_name

        and

        "id"
        in
        lowered_name
    ):

        continue


    function_code = (
        getattr(
            value,
            "__code__",
            None,
        )
    )


    if (
        function_code
        is None
    ):

        continue


    argument_names = list(
        function_code.co_varnames[
            :
            function_code.co_argcount
        ]
    )


    cell82_question_id_builder_candidates.append(
        {
            "name":
                name,

            "callable":
                value,

            "argument_names":
                argument_names,
        }
    )


CELL82_REVISION = 1


cell82_id_argument_values = {
    "pilot_case_id":
        cell82_packet[
            "pilot_case_id"
        ],

    "golden_case_id":
        cell82_packet[
            "golden_case_id"
        ],

    "task_draft_id":
        cell82_packet[
            "task_draft_id"
        ],

    "target_evidence_id":
        cell82_packet[
            "target_evidence_id"
        ],

    "question_text":
        CELL82_QUESTION_TEXT,

    "question_text_sha256":
        cell82_question_text_sha256,

    "revision":
        CELL82_REVISION,

    "revision_number":
        CELL82_REVISION,

    "question_revision":
        CELL82_REVISION,

    "draft_revision":
        CELL82_REVISION,

    "schema_version":
        cell82_schema_version,

    "question_schema_version":
        cell82_schema_version,

    "question_draft_schema_version":
        cell82_schema_version,
}


cell82_compatible_id_builders = []


for candidate in (
    cell82_question_id_builder_candidates
):

    if all(
        argument_name
        in
        cell82_id_argument_values

        for argument_name
        in candidate[
            "argument_names"
        ]
    ):

        cell82_compatible_id_builders.append(
            candidate
        )


def cell82_builder_priority(candidate):

    name = (
        candidate[
            "name"
        ].lower()
    )

    if (
        name
        ==
        "build_question_draft_id"
    ):

        return (
            0,
            name,
        )


    if (
        "build_question_draft_id"
        in
        name
    ):

        return (
            1,
            name,
        )


    if (
        "question_draft"
        in
        name
    ):

        return (
            2,
            name,
        )


    return (
        3,
        name,
    )


cell82_compatible_id_builders = sorted(
    cell82_compatible_id_builders,
    key=cell82_builder_priority,
)


cell82_id_failures = []


if (
    len(
        cell82_compatible_id_builders
    )
    ==
    0
):

    cell82_id_failures.append(
        {
            "reason":
                "NO_COMPATIBLE_FROZEN_QUESTION_ID_BUILDER_FOUND",

            "discovered":
                [
                    {
                        "name":
                            row[
                                "name"
                            ],

                        "argument_names":
                            row[
                                "argument_names"
                            ],
                    }

                    for row
                    in cell82_question_id_builder_candidates
                ],
        }
    )


cell82_selected_id_builder = (
    cell82_compatible_id_builders[
        0
    ]

    if (
        len(
            cell82_compatible_id_builders
        )
        >
        0
    )

    else None
)


cell82_question_draft_id = None


if (
    cell82_selected_id_builder
    is not None
):

    cell82_builder_kwargs = {
        argument_name:
            cell82_id_argument_values[
                argument_name
            ]

        for argument_name
        in cell82_selected_id_builder[
            "argument_names"
        ]
    }


    try:

        cell82_question_draft_id = (
            cell82_selected_id_builder[
                "callable"
            ](
                **cell82_builder_kwargs
            )
        )

    except Exception as exc:

        cell82_id_failures.append(
            {
                "reason":
                    "QUESTION_ID_BUILDER_EXECUTION_FAILED",

                "builder":
                    cell82_selected_id_builder[
                        "name"
                    ],

                "error_type":
                    type(
                        exc
                    ).__name__,

                "error_message":
                    str(
                        exc
                    ),
            }
        )


if not (
    isinstance(
        cell82_question_draft_id,
        str,
    )

    and

    cell82_question_draft_id.startswith(
        "question-draft:"
    )
):

    cell82_id_failures.append(
        {
            "reason":
                "QUESTION_DRAFT_ID_INVALID",

            "actual":
                cell82_question_draft_id,
        }
    )


# ============================================================
# 8) Determinism check
# ============================================================

cell82_question_draft_id_repeat = None


if (
    cell82_selected_id_builder
    is not None

    and

    len(
        cell82_id_failures
    )
    ==
    0
):

    cell82_question_draft_id_repeat = (
        cell82_selected_id_builder[
            "callable"
        ](
            **cell82_builder_kwargs
        )
    )


cell82_id_deterministic = (
    cell82_question_draft_id
    is not None

    and

    cell82_question_draft_id
    ==
    cell82_question_draft_id_repeat
)


# ============================================================
# 9) Frozen initial status contract
# ============================================================

CELL82_INITIAL_STATUSES = {
    "draft_status":
        "DRAFT_CREATED_PENDING_VALIDATION",

    "machine_validation_status":
        "NOT_STARTED",

    "human_adjudication_status":
        "NOT_STARTED",

    "expected_answer_status":
        "BLOCKED_PENDING_QUESTION_ACCEPTANCE",

    "citation_target_validation_status":
        "BLOCKED_PENDING_QUESTION_ACCEPTANCE",
}


# ============================================================
# 10) Candidate field universe
# ============================================================

cell82_question_field_values = {
    "question_draft_id":
        cell82_question_draft_id,

    "schema_version":
        cell82_schema_version,

    "question_schema_version":
        cell82_schema_version,

    "question_draft_schema_version":
        cell82_schema_version,

    "question_draft_version":
        cell82_schema_version,

    "pilot_case_id":
        cell82_packet[
            "pilot_case_id"
        ],

    "golden_case_id":
        cell82_packet[
            "golden_case_id"
        ],

    "task_draft_id":
        cell82_packet[
            "task_draft_id"
        ],

    "target_evidence_id":
        cell82_packet[
            "target_evidence_id"
        ],

    "source_type":
        cell82_packet[
            "source_type"
        ],

    "stratum":
        cell82_packet[
            "stratum"
        ],

    "task_family":
        cell82_packet[
            "task_family"
        ],

    "question_archetype":
        cell82_packet[
            "question_archetype"
        ],

    "required_dimensions":
        list(
            cell82_packet[
                "required_dimensions"
            ]
        ),

    "authority_expectation":
        cell82_packet[
            "authority_expectation"
        ],

    "context_expectation":
        cell82_packet[
            "context_expectation"
        ],

    "evidence_scope":
        cell82_packet[
            "evidence_scope"
        ],

    "target_selection_method":
        cell82_packet[
            "target_selection_method"
        ],

    "target_selection_basis":
        cell82_packet[
            "target_selection_basis"
        ],

    "target_text_parts":
        list(
            cell82_packet[
                "target_text_parts"
            ]
        ),

    "guardrails":
        dict(
            cell82_packet[
                "guardrails"
            ]
        ),

    "authoring_mode":
        PILOT_QUESTION_AUTHORING_MODE_V1[
            "mode"
        ],

    "authoring_packet_sha256":
        cell81_authoring_packet_sha256,

    "question_text":
        CELL82_QUESTION_TEXT,

    "question_text_sha256":
        cell82_question_text_sha256,

    "revision":
        CELL82_REVISION,

    "revision_number":
        CELL82_REVISION,

    "question_revision":
        CELL82_REVISION,

    "draft_revision":
        CELL82_REVISION,

    "draft_status":
        CELL82_INITIAL_STATUSES[
            "draft_status"
        ],

    "question_draft_status":
        CELL82_INITIAL_STATUSES[
            "draft_status"
        ],

    "machine_validation_status":
        CELL82_INITIAL_STATUSES[
            "machine_validation_status"
        ],

    "question_groundability_status":
        "NOT_STARTED",

    "human_adjudication_status":
        CELL82_INITIAL_STATUSES[
            "human_adjudication_status"
        ],

    "expected_answer_status":
        CELL82_INITIAL_STATUSES[
            "expected_answer_status"
        ],

    "citation_target_validation_status":
        CELL82_INITIAL_STATUSES[
            "citation_target_validation_status"
        ],

    "expected_answer_generation_status":
        "BLOCKED_PENDING_QUESTION_ACCEPTANCE",

    "citation_target_status":
        "BLOCKED_PENDING_QUESTION_ACCEPTANCE",
}


cell82_unknown_required_fields = [
    field_name

    for field_name
    in cell82_schema_required_fields

    if (
        field_name
        not in
        cell82_question_field_values
    )
]


if (
    cell82_unknown_required_fields
):

    cell82_schema_failures.append(
        {
            "reason":
                "UNMAPPED_REQUIRED_QUESTION_SCHEMA_FIELDS",

            "fields":
                cell82_unknown_required_fields,
        }
    )


# ============================================================
# 11) Materialize exact schema-required record
# ============================================================

cell82_first_question_draft_record = (
    {
        field_name:
            cell82_question_field_values[
                field_name
            ]

        for field_name
        in cell82_schema_required_fields
    }

    if (
        len(
            cell82_schema_failures
        )
        ==
        0

        and

        len(
            cell82_id_failures
        )
        ==
        0
    )

    else None
)


cell82_record_failures = []


if (
    cell82_first_question_draft_record
    is None
):

    cell82_record_failures.append(
        "QUESTION_DRAFT_RECORD_NOT_MATERIALIZED"
    )

else:

    missing_fields = [
        field_name

        for field_name
        in cell82_schema_required_fields

        if (
            field_name
            not in
            cell82_first_question_draft_record
        )
    ]


    if (
        missing_fields
    ):

        cell82_record_failures.append(
            {
                "reason":
                    "QUESTION_DRAFT_REQUIRED_FIELDS_MISSING",

                "fields":
                    missing_fields,
            }
        )


cell82_question_draft_record_sha256 = (
    cell82_sha(
        cell82_first_question_draft_record
    )

    if (
        cell82_first_question_draft_record
        is not None
    )

    else None
)


# ============================================================
# 12) Freeze state after drafting
# ============================================================

cell82_active_snapshot_after = (
    cell82_sha(
        {
            name:
                globals()[
                    name
                ]

            for name
            in CELL82_ACTIVE_STATE_OBJECT_NAMES

            if (
                name
                in
                globals()
            )
        }
    )
)


cell82_pilot_snapshot_after = (
    cell82_sha(
        {
            "question_drafting_pilot_packets":
                question_drafting_pilot_packets,

            "question_drafting_pilot_by_id":
                question_drafting_pilot_by_id,
        }
    )
)


# ============================================================
# 13) Gates
# ============================================================

cell82_gates = {
    "cell81_is_pass":
        (
            globals().get(
                "cell_81_pass"
            )
            is True
        ),

    "first_authoring_packet_is_exact_expected_pilot":
        (
            len(
                cell82_packet_failures
            )
            ==
            0
        ),

    "question_is_natural_nonempty_and_evidence_bounded":
        (
            len(
                cell82_preflight_failures
            )
            ==
            0
        ),

    "question_schema_is_frozen_v1":
        (
            cell82_schema_version
            ==
            "question_draft_record_v1"
        ),

    "all_schema_required_fields_are_mapped":
        (
            len(
                cell82_schema_failures
            )
            ==
            0
        ),

    "frozen_question_id_builder_is_resolved":
        (
            cell82_selected_id_builder
            is not None

            and

            len(
                cell82_id_failures
            )
            ==
            0
        ),

    "question_draft_id_is_deterministic":
        cell82_id_deterministic,

    "question_draft_record_is_materialized":
        (
            cell82_first_question_draft_record
            is not None

            and

            len(
                cell82_record_failures
            )
            ==
            0
        ),

    "human_adjudication_not_started":
        (
            CELL82_INITIAL_STATUSES[
                "human_adjudication_status"
            ]
            ==
            "NOT_STARTED"
        ),

    "expected_answer_remains_blocked":
        (
            CELL82_INITIAL_STATUSES[
                "expected_answer_status"
            ]
            ==
            "BLOCKED_PENDING_QUESTION_ACCEPTANCE"
        ),

    "citation_validation_remains_blocked":
        (
            CELL82_INITIAL_STATUSES[
                "citation_target_validation_status"
            ]
            ==
            "BLOCKED_PENDING_QUESTION_ACCEPTANCE"
        ),

    "active_state_remains_byte_stable":
        (
            cell82_active_snapshot_before
            ==
            cell82_active_snapshot_after
        ),

    "pilot_runtime_remains_byte_stable":
        (
            cell82_pilot_snapshot_before
            ==
            cell82_pilot_snapshot_after
        ),

    "no_artifact_write":
        True,
}


cell_82_pass = all(
    cell82_gates.values()
)


cell82_status = (
    "FIRST_QUESTION_DRAFT_MATERIALIZED_PENDING_MACHINE_VALIDATION"

    if (
        cell_82_pass
    )

    else
    "FIRST_QUESTION_DRAFT_MATERIALIZATION_FAILED"
)


# ============================================================
# 14) Diagnostics
# ============================================================

print(
    "=== FIRST REAL QUESTION DRAFT MATERIALIZATION ==="
)


print(
    "\n=== PREREQUISITES ==="
)

print(
    "Failures:",
    cell82_prerequisite_failures
)


print(
    "\n=== QUESTION ==="
)

print(
    CELL82_QUESTION_TEXT
)

print(
    "Question SHA256:",
    cell82_question_text_sha256
)


print(
    "\n=== AUTHORING PREFLIGHT ==="
)

for gate_name, gate_value in (
    cell82_question_preflight.items()
):

    print(
        gate_name,
        ":",
        (
            "PASS"
            if gate_value
            else "FAIL"
        )
    )


print(
    "Preflight failures:",
    cell82_preflight_failures
)


print(
    "\n=== FROZEN SCHEMA ==="
)

print(
    "Schema version:",
    cell82_schema_version
)

print(
    "Required fields:",
    cell82_schema_required_fields
)

print(
    "Schema failures:",
    cell82_schema_failures
)


print(
    "\n=== QUESTION ID BUILDER ==="
)

print(
    "Discovered:",
    [
        {
            "name":
                row[
                    "name"
                ],

            "argument_names":
                row[
                    "argument_names"
                ],
        }

        for row
        in cell82_question_id_builder_candidates
    ]
)

print(
    "Selected:",
    (
        cell82_selected_id_builder[
            "name"
        ]

        if (
            cell82_selected_id_builder
            is not None
        )

        else None
    )
)

print(
    "Question Draft ID:",
    cell82_question_draft_id
)

print(
    "Repeat ID:",
    cell82_question_draft_id_repeat
)

print(
    "ID failures:",
    cell82_id_failures
)


print(
    "\n=== DRAFT RECORD ==="
)

print(
    "Materialized:",
    (
        cell82_first_question_draft_record
        is not None
    )
)

print(
    "Record fields:",
    (
        list(
            cell82_first_question_draft_record.keys()
        )

        if (
            cell82_first_question_draft_record
            is not None
        )

        else []
    )
)

print(
    "Record SHA256:",
    cell82_question_draft_record_sha256
)

print(
    "Record failures:",
    cell82_record_failures
)


print(
    "\n=== BLOCKED DOWNSTREAM STAGES ==="
)

print(
    "Machine validation:",
    "NOT_STARTED"
)

print(
    "Human adjudication:",
    "NOT_STARTED"
)

print(
    "Expected answer:",
    "BLOCKED_PENDING_QUESTION_ACCEPTANCE"
)

print(
    "Citation target validation:",
    "BLOCKED_PENDING_QUESTION_ACCEPTANCE"
)

print(
    "Full 572 generation:",
    "BLOCKED"
)


print(
    "\nSTATUS:",
    cell82_status
)


print(
    "\n=== CODE CELL 82 GATES ==="
)


for gate_name, gate_value in (
    cell82_gates.items()
):

    print(
        gate_name,
        ":",
        (
            "PASS"
            if gate_value
            else "FAIL"
        )
    )


print(
    "\nFAILED GATES:",
    [
        gate_name

        for gate_name, gate_value
        in cell82_gates.items()

        if not gate_value
    ]
)


print(
    "\nCODE CELL 82 GATE:",
    (
        "PASS"
        if cell_82_pass
        else "FAIL"
    )
)


assert cell_82_pass, (
    "First Question Draft materialization failed. "
    "Do not run Machine Validation or Human Adjudication. "
    "Repair this SAME Code Cell 82 using the printed diagnostics."
)
```

<h3>Machine Validation اولین Question Draft</h3>

<p>
Cell 82 اولین Question Draft واقعی را با Schema و ID پایدار Materialize کرد، اما وضعیت آن هنوز
<code>DRAFT_CREATED_PENDING_VALIDATION</code>

</p>

<p>
در این مرحله همان Draft بر اساس
<code>QUESTION_DRAFT_ACCEPTANCE_RUBRIC_V1</code>
به‌صورت Machine Validation بررسی می‌شود.
</p>

<h4>دامنه‌ی Validation</h4>

<p>
دو گروه Gate باید مستقل بررسی شوند:
</p>

<ul>
<li><strong>Hard Gates</strong>: Groundability، نبود External Knowledge، نبود Answer Leakage، نبود Internal Metadata، صحت Attribution و Authority، نبود Locator Fabrication و معتبر بودن Target Evidence.</li>
<li><strong>Semantic Gates</strong>: هم‌راستایی سؤال با Task Family، Question Archetype، Evidence Scope، Authority، Context و Required Dimensions.</li>
</ul>

<p>
Machine Validation اجازه ندارد Question را Accepted اعلام کند. حتی در صورت PASS کامل، تصمیم نهایی Human Adjudication همچنان اجباری است.
</p>

<h4>مرز این مرحله</h4>

<ul>
<li>Question Text تغییر نمی‌کند؛</li>
<li>Human Decision انجام نمی‌شود؛</li>
<li>Expected Answer تولید نمی‌شود؛</li>
<li>Citation Target Validation انجام نمی‌شود؛</li>
<li>هیچ Active Dataset یا Pilot Packet تغییر نمی‌کند؛</li>
<li>هیچ Artifact روی Disk نوشته نمی‌شود؛</li>
<li>Generation برای 572 Case همچنان Blocked می‌ماند.</li>
</ul>

<p>
اگر تمام Hard Gateها و Semantic Gateها PASS شوند، Draft فقط به وضعیت
<code>MACHINE_VALIDATION_PASS_PENDING_HUMAN_ADJUDICATION</code>
می‌رسد.
</p>

</div>

```python
# Code Cell 83

# ============================================================
# Machine Validation of the first real Question Draft
#
# No human acceptance.
# No expected-answer generation.
# No citation adjudication.
# No active-state mutation.
# No artifact write.
# ============================================================


# ============================================================
# 1) Prerequisites
# ============================================================

cell83_prerequisite_failures = []


if (
    globals().get(
        "cell_82_pass"
    )
    is not True
):

    cell83_prerequisite_failures.append(
        "CELL82_NOT_PASS"
    )


if (
    globals().get(
        "cell82_status"
    )
    !=
    "FIRST_QUESTION_DRAFT_MATERIALIZED_PENDING_MACHINE_VALIDATION"
):

    cell83_prerequisite_failures.append(
        {
            "reason":
                "CELL82_STATUS_UNEXPECTED",

            "actual":
                globals().get(
                    "cell82_status"
                ),
        }
    )


CELL83_REQUIRED_RUNTIME_OBJECTS = [
    "cell82_first_question_draft_record",
    "cell82_question_draft_record_sha256",
    "cell82_packet",
    "cell81_selected_task",
    "QUESTION_DRAFT_ACCEPTANCE_RUBRIC_V1",
]


cell83_missing_runtime_objects = [
    name

    for name
    in CELL83_REQUIRED_RUNTIME_OBJECTS

    if (
        name
        not in
        globals()
    )
]


if (
    cell83_missing_runtime_objects
):

    cell83_prerequisite_failures.append(
        {
            "reason":
                "REQUIRED_RUNTIME_OBJECTS_MISSING",

            "objects":
                cell83_missing_runtime_objects,
        }
    )


# ============================================================
# 2) Helpers
# ============================================================

def cell83_sha(value):

    return hashlib.sha256(
        stable_json_key(
            value
        ).encode(
            "utf-8"
        )
    ).hexdigest()


# ============================================================
# 3) Freeze state BEFORE validation
# ============================================================

CELL83_ACTIVE_STATE_OBJECT_NAMES = [
    "golden_case_records",
    "case_groundability_precheck",
    "case_task_design",
    "task_draft_plan_rows",
    "target_evidence_records",
    "pilot_selection_rows",
    "golden_case_records_by_id",
    "task_draft_plans",
    "target_evidence_by_case_id",
    "target_evidence_by_task_draft_id",
]


cell83_active_snapshot_before = (
    cell83_sha(
        {
            name:
                globals()[
                    name
                ]

            for name
            in CELL83_ACTIVE_STATE_OBJECT_NAMES

            if (
                name
                in
                globals()
            )
        }
    )
)


cell83_pilot_snapshot_before = (
    cell83_sha(
        {
            "question_drafting_pilot_packets":
                question_drafting_pilot_packets,

            "question_drafting_pilot_by_id":
                question_drafting_pilot_by_id,
        }
    )
)


cell83_question_record_snapshot_before = (
    cell83_sha(
        cell82_first_question_draft_record
    )
)


# ============================================================
# 4) Resolve frozen Rubric
# ============================================================

cell83_rubric = (
    QUESTION_DRAFT_ACCEPTANCE_RUBRIC_V1
)


cell83_hard_gate_contract = (
    cell83_rubric.get(
        "hard_gates",
        {},
    )
)


cell83_semantic_gate_contract = (
    cell83_rubric.get(
        "semantic_gates",
        {},
    )
)


CELL83_EXPECTED_HARD_GATES = {
    "nonempty_question_text",
    "natural_user_question",
    "external_knowledge_dependency",
    "answer_leakage_in_question",
    "internal_document_id_leakage",
    "internal_pipeline_metadata_leakage",
    "unknown_speaker_inference",
    "unsupported_attribution",
    "unsupported_authority_promotion",
    "locator_fabrication",
    "question_must_be_groundable_in_target_text_parts",
    "target_evidence_must_already_be_validated",
}


CELL83_EXPECTED_SEMANTIC_GATES = {
    "task_family_alignment",
    "question_archetype_alignment",
    "evidence_scope_alignment",
    "authority_expectation_alignment",
    "context_expectation_alignment",
    "required_dimensions_are_testable_by_question",
    "question_is_not_trivial_metadata_repetition",
    "question_is_answerable_without_external_context",
}


cell83_rubric_failures = []


if (
    cell83_rubric.get(
        "version"
    )
    !=
    "question_draft_acceptance_rubric_v1"
):

    cell83_rubric_failures.append(
        {
            "reason":
                "RUBRIC_VERSION_UNEXPECTED",

            "actual":
                cell83_rubric.get(
                    "version"
                ),
        }
    )


if (
    set(
        cell83_hard_gate_contract.keys()
    )
    !=
    CELL83_EXPECTED_HARD_GATES
):

    cell83_rubric_failures.append(
        {
            "reason":
                "HARD_GATE_CONTRACT_NOT_EXACT",

            "actual":
                sorted(
                    cell83_hard_gate_contract.keys()
                ),
        }
    )


if (
    set(
        cell83_semantic_gate_contract.keys()
    )
    !=
    CELL83_EXPECTED_SEMANTIC_GATES
):

    cell83_rubric_failures.append(
        {
            "reason":
                "SEMANTIC_GATE_CONTRACT_NOT_EXACT",

            "actual":
                sorted(
                    cell83_semantic_gate_contract.keys()
                ),
        }
    )


if (
    cell83_rubric.get(
        "human_review_policy",
        {},
    ).get(
        "required_before_acceptance"
    )
    is not True
):

    cell83_rubric_failures.append(
        "HUMAN_REVIEW_NOT_REQUIRED_BY_RUBRIC"
    )


# ============================================================
# 5) Resolve Question and Evidence
# ============================================================

cell83_question_record = (
    cell82_first_question_draft_record
)


cell83_question_text = (
    cell83_question_record.get(
        "question_text",
        "",
    )
)


cell83_target_parts = (
    cell82_packet.get(
        "target_text_parts",
        [],
    )
)


cell83_joined_evidence = "\n".join(
    part.get(
        "text",
        "",
    )

    for part
    in cell83_target_parts
)


cell83_attribution_names = set()

cell83_valid_direct_author_parts = []


for part in (
    cell83_target_parts
):

    attribution = (
        part.get(
            "attribution"
        )
        or
        {}
    )


    for field_name in [
        "author",
        "credited_author_name",
        "knowledge_owner",
    ]:

        value = (
            attribution.get(
                field_name
            )
        )


        if (
            isinstance(
                value,
                str,
            )

            and

            value.strip()
        ):

            cell83_attribution_names.add(
                value.strip()
            )


    if (
        attribution.get(
            "attribution_status"
        )
        ==
        "policy_validated"

        and

        attribution.get(
            "content_authority_type"
        )
        ==
        "direct_author"
    ):

        cell83_valid_direct_author_parts.append(
            part
        )


# ============================================================
# 6) Evidence facts used by validation
#
# These are validation anchors, not an Expected Answer record.
# ============================================================

CELL83_REQUIRED_ANSWER_EVIDENCE_ANCHORS = [
    "سامانه سجام",
    "احراز هویت",
    "کد معاملاتی",
    "کاغذی",
    "تلفنی",
    "اینترنتی",
]


cell83_missing_evidence_anchors = [
    anchor

    for anchor
    in CELL83_REQUIRED_ANSWER_EVIDENCE_ANCHORS

    if (
        anchor
        not in
        cell83_joined_evidence
    )
]


CELL83_ANSWER_LEAKAGE_ANCHORS = [
    "سامانه سجام",
    "احراز هویت",
    "کد معاملاتی",
    "روش کاغذی",
    "روش تلفنی",
    "روش اینترنتی",
]


cell83_answer_leakage_hits = [
    anchor

    for anchor
    in CELL83_ANSWER_LEAKAGE_ANCHORS

    if (
        anchor
        in
        cell83_question_text
    )
]


CELL83_INTERNAL_METADATA_MARKERS = [
    "golden-case:",
    "task-draft:",
    "target-evidence:",
    "question-pilot:",
    "question-draft:",
    "sha256",
    "candidate_identity",
    "document_id",
    "source_id",
]


cell83_internal_metadata_hits = [
    marker

    for marker
    in CELL83_INTERNAL_METADATA_MARKERS

    if (
        marker
        in
        cell83_question_text
    )
]


CELL83_LOCATOR_MARKERS = [
    "http://",
    "https://",
    "www.",
    "صفحه ",
    "خط ",
    "فایل ",
]


cell83_locator_hits = [
    marker

    for marker
    in CELL83_LOCATOR_MARKERS

    if (
        marker
        in
        cell83_question_text
    )
]


# ============================================================
# 7) Machine Validation — Hard Gates
# ============================================================

cell83_hard_gate_results = {
    "nonempty_question_text":
        (
            isinstance(
                cell83_question_text,
                str,
            )

            and

            len(
                cell83_question_text.strip()
            )
            >
            0
        ),

    "natural_user_question":
        (
            isinstance(
                cell83_question_text,
                str,
            )

            and

            len(
                cell83_question_text.strip()
            )
            >=
            20

            and

            cell83_question_text.strip().endswith(
                "؟"
            )

            and

            "\n"
            not in
            cell83_question_text.strip()
        ),

    "external_knowledge_dependency":
        (
            len(
                cell83_missing_evidence_anchors
            )
            ==
            0
        ),

    "answer_leakage_in_question":
        (
            len(
                cell83_answer_leakage_hits
            )
            ==
            0
        ),

    "internal_document_id_leakage":
        (
            not any(
                marker
                in
                cell83_question_text

                for marker
                in [
                    "document_id",
                    "source_id",
                    "golden-case:",
                    "task-draft:",
                    "target-evidence:",
                    "question-pilot:",
                ]
            )
        ),

    "internal_pipeline_metadata_leakage":
        (
            len(
                cell83_internal_metadata_hits
            )
            ==
            0
        ),

    "unknown_speaker_inference":
        (
            "گوینده"
            not in
            cell83_question_text

            and

            "speaker"
            not in
            cell83_question_text.lower()
        ),

    "unsupported_attribution":
        (
            "دکتر کمیل رودی"
            in
            cell83_attribution_names
        ),

    "unsupported_authority_promotion":
        (
            cell82_packet.get(
                "authority_expectation"
            )
            ==
            "PRIMARY_DIRECT_AUTHOR"

            and

            len(
                cell83_valid_direct_author_parts
            )
            >=
            1
        ),

    "locator_fabrication":
        (
            len(
                cell83_locator_hits
            )
            ==
            0
        ),

    "question_must_be_groundable_in_target_text_parts":
        (
            len(
                cell83_target_parts
            )
            >=
            1

            and

            len(
                cell83_missing_evidence_anchors
            )
            ==
            0
        ),

    "target_evidence_must_already_be_validated":
        (
            cell81_selected_task.get(
                "target_evidence_status"
            )
            ==
            "SELECTED_AND_VALIDATED"
        ),
}


# ============================================================
# 8) Required-dimension testability
# ============================================================

cell83_required_dimensions = (
    cell82_packet.get(
        "required_dimensions",
        [],
    )
)


CELL83_DIMENSION_TESTABILITY = {
    "retrieval_hit":
        (
            "Question requires retrieval of explicit factual "
            "content from the target document."
        ),

    "rank":
        (
            "The target document can be evaluated for ranking "
            "relative to competing retrieval candidates."
        ),

    "attribution_correctness":
        (
            "Question explicitly names the validated direct author."
        ),

    "citation_locator_correctness":
        (
            "A returned answer can be checked against the exact "
            "source-document locator carried by the target part."
        ),

    "source_authority_correctness":
        (
            "Question explicitly constrains the answer to the "
            "validated Primary Direct Author source."
        ),
}


cell83_uncovered_required_dimensions = [
    dimension

    for dimension
    in cell83_required_dimensions

    if (
        dimension
        not in
        CELL83_DIMENSION_TESTABILITY
    )
]


# ============================================================
# 9) Machine semantic adjudication
#
# Explicit model-authored semantic decisions.
# These do NOT constitute Human Adjudication.
# ============================================================

CELL83_SEMANTIC_GATE_JUDGMENTS = {
    "task_family_alignment": {
        "status":
            "PASS",

        "reason":
            (
                "The question asks for factual retrieval directly "
                "supported by the selected Primary source document."
            ),
    },

    "question_archetype_alignment": {
        "status":
            "PASS",

        "reason":
            (
                "The phrase 'بر اساس توضیح دکتر کمیل رودی' makes "
                "the authority constraint explicit."
            ),
    },

    "evidence_scope_alignment": {
        "status":
            "PASS",

        "reason":
            (
                "All information necessary to answer the question "
                "is contained in the single selected source document."
            ),
    },

    "authority_expectation_alignment": {
        "status":
            "PASS",

        "reason":
            (
                "The named author is policy-validated and the "
                "content authority type is direct_author."
            ),
    },

    "context_expectation_alignment": {
        "status":
            "PASS",

        "reason":
            (
                "The question requires only the minimal direct "
                "evidence contained in the selected target part."
            ),
    },

    "required_dimensions_are_testable_by_question": {
        "status":
            (
                "PASS"

                if (
                    len(
                        cell83_uncovered_required_dimensions
                    )
                    ==
                    0
                )

                else
                "FAIL"
            ),

        "reason":
            (
                "Every packet-required evaluation dimension has "
                "an explicit testability rationale."
            ),
    },

    "question_is_not_trivial_metadata_repetition": {
        "status":
            "PASS",

        "reason":
            (
                "The question requests substantive source content "
                "rather than IDs, metadata, title fields, or locator text."
            ),
    },

    "question_is_answerable_without_external_context": {
        "status":
            (
                "PASS"

                if (
                    len(
                        cell83_missing_evidence_anchors
                    )
                    ==
                    0
                )

                else
                "FAIL"
            ),

        "reason":
            (
                "The target evidence contains all required answer "
                "components; no outside information is necessary."
            ),
    },
}


cell83_semantic_gate_results = {
    gate_name:
        (
            judgment.get(
                "status"
            )
            ==
            "PASS"
        )

    for gate_name, judgment
    in CELL83_SEMANTIC_GATE_JUDGMENTS.items()
}


# ============================================================
# 10) Contract/result coverage validation
# ============================================================

cell83_validation_structure_failures = []


if (
    set(
        cell83_hard_gate_results.keys()
    )
    !=
    set(
        cell83_hard_gate_contract.keys()
    )
):

    cell83_validation_structure_failures.append(
        {
            "reason":
                "HARD_GATE_RESULT_COVERAGE_NOT_EXACT",

            "expected":
                sorted(
                    cell83_hard_gate_contract.keys()
                ),

            "actual":
                sorted(
                    cell83_hard_gate_results.keys()
                ),
        }
    )


if (
    set(
        cell83_semantic_gate_results.keys()
    )
    !=
    set(
        cell83_semantic_gate_contract.keys()
    )
):

    cell83_validation_structure_failures.append(
        {
            "reason":
                "SEMANTIC_GATE_RESULT_COVERAGE_NOT_EXACT",

            "expected":
                sorted(
                    cell83_semantic_gate_contract.keys()
                ),

            "actual":
                sorted(
                    cell83_semantic_gate_results.keys()
                ),
        }
    )


# ============================================================
# 11) Machine decision
# ============================================================

cell83_failed_hard_gates = [
    gate_name

    for gate_name, result
    in cell83_hard_gate_results.items()

    if not result
]


cell83_failed_semantic_gates = [
    gate_name

    for gate_name, result
    in cell83_semantic_gate_results.items()

    if not result
]


cell83_machine_validation_decision = (
    "PASS"

    if (
        len(
            cell83_prerequisite_failures
        )
        ==
        0

        and

        len(
            cell83_rubric_failures
        )
        ==
        0

        and

        len(
            cell83_validation_structure_failures
        )
        ==
        0

        and

        len(
            cell83_failed_hard_gates
        )
        ==
        0

        and

        len(
            cell83_failed_semantic_gates
        )
        ==
        0
    )

    else
    "FAIL"
)


# ============================================================
# 12) Materialize read-only Machine Validation record
# ============================================================

cell83_machine_validation_record = {
    "validation_version":
        "question_draft_machine_validation_v1",

    "question_draft_id":
        cell83_question_record[
            "question_draft_id"
        ],

    "pilot_case_id":
        cell83_question_record[
            "pilot_case_id"
        ],

    "question_text_sha256":
        cell83_question_record[
            "question_text_sha256"
        ],

    "question_draft_record_sha256":
        cell82_question_draft_record_sha256,

    "rubric_version":
        cell83_rubric[
            "version"
        ],

    "hard_gate_results":
        dict(
            cell83_hard_gate_results
        ),

    "semantic_gate_results":
        dict(
            cell83_semantic_gate_results
        ),

    "semantic_gate_judgments":
        dict(
            CELL83_SEMANTIC_GATE_JUDGMENTS
        ),

    "required_dimension_testability":
        {
            dimension:
                CELL83_DIMENSION_TESTABILITY[
                    dimension
                ]

            for dimension
            in cell83_required_dimensions

            if (
                dimension
                in
                CELL83_DIMENSION_TESTABILITY
            )
        },

    "machine_validation_decision":
        cell83_machine_validation_decision,

    "human_adjudication_required":
        True,

    "human_adjudication_status":
        "NOT_STARTED",

    "automatic_acceptance":
        False,

    "expected_answer_status":
        "BLOCKED_PENDING_QUESTION_ACCEPTANCE",

    "citation_target_validation_status":
        "BLOCKED_PENDING_QUESTION_ACCEPTANCE",

    "full_572_generation":
        "BLOCKED",
}


cell83_machine_validation_record_sha256 = (
    cell83_sha(
        cell83_machine_validation_record
    )
)


# ============================================================
# 13) Freeze state AFTER validation
# ============================================================

cell83_active_snapshot_after = (
    cell83_sha(
        {
            name:
                globals()[
                    name
                ]

            for name
            in CELL83_ACTIVE_STATE_OBJECT_NAMES

            if (
                name
                in
                globals()
            )
        }
    )
)


cell83_pilot_snapshot_after = (
    cell83_sha(
        {
            "question_drafting_pilot_packets":
                question_drafting_pilot_packets,

            "question_drafting_pilot_by_id":
                question_drafting_pilot_by_id,
        }
    )
)


cell83_question_record_snapshot_after = (
    cell83_sha(
        cell82_first_question_draft_record
    )
)


# ============================================================
# 14) Final gates
# ============================================================

cell83_gates = {
    "cell82_is_pass":
        (
            globals().get(
                "cell_82_pass"
            )
            is True
        ),

    "cell82_status_is_pending_machine_validation":
        (
            globals().get(
                "cell82_status"
            )
            ==
            "FIRST_QUESTION_DRAFT_MATERIALIZED_PENDING_MACHINE_VALIDATION"
        ),

    "frozen_rubric_is_exact_and_available":
        (
            len(
                cell83_rubric_failures
            )
            ==
            0
        ),

    "machine_validation_covers_every_hard_gate":
        (
            set(
                cell83_hard_gate_results.keys()
            )
            ==
            set(
                cell83_hard_gate_contract.keys()
            )
        ),

    "all_hard_gates_pass":
        (
            len(
                cell83_failed_hard_gates
            )
            ==
            0
        ),

    "machine_validation_covers_every_semantic_gate":
        (
            set(
                cell83_semantic_gate_results.keys()
            )
            ==
            set(
                cell83_semantic_gate_contract.keys()
            )
        ),

    "all_semantic_gates_pass":
        (
            len(
                cell83_failed_semantic_gates
            )
            ==
            0
        ),

    "all_required_dimensions_have_testability_rationale":
        (
            len(
                cell83_uncovered_required_dimensions
            )
            ==
            0
        ),

    "machine_validation_decision_is_pass":
        (
            cell83_machine_validation_decision
            ==
            "PASS"
        ),

    "human_adjudication_is_still_required":
        (
            cell83_machine_validation_record[
                "human_adjudication_required"
            ]
            is True

            and

            cell83_machine_validation_record[
                "human_adjudication_status"
            ]
            ==
            "NOT_STARTED"

            and

            cell83_machine_validation_record[
                "automatic_acceptance"
            ]
            is False
        ),

    "expected_answer_remains_blocked":
        (
            cell83_machine_validation_record[
                "expected_answer_status"
            ]
            ==
            "BLOCKED_PENDING_QUESTION_ACCEPTANCE"
        ),

    "citation_validation_remains_blocked":
        (
            cell83_machine_validation_record[
                "citation_target_validation_status"
            ]
            ==
            "BLOCKED_PENDING_QUESTION_ACCEPTANCE"
        ),

    "full_572_generation_remains_blocked":
        (
            cell83_machine_validation_record[
                "full_572_generation"
            ]
            ==
            "BLOCKED"
        ),

    "question_draft_record_remains_byte_stable":
        (
            cell83_question_record_snapshot_before
            ==
            cell83_question_record_snapshot_after
        ),

    "active_state_remains_byte_stable":
        (
            cell83_active_snapshot_before
            ==
            cell83_active_snapshot_after
        ),

    "pilot_runtime_remains_byte_stable":
        (
            cell83_pilot_snapshot_before
            ==
            cell83_pilot_snapshot_after
        ),

    "no_artifact_write":
        True,
}


cell_83_pass = all(
    cell83_gates.values()
)


cell83_status = (
    "MACHINE_VALIDATION_PASS_PENDING_HUMAN_ADJUDICATION"

    if (
        cell_83_pass
    )

    else
    "MACHINE_VALIDATION_FAILED"
)


# ============================================================
# 15) Diagnostics
# ============================================================

print(
    "=== FIRST QUESTION DRAFT — MACHINE VALIDATION ==="
)


print(
    "\n=== PREREQUISITES ==="
)

print(
    "Failures:",
    cell83_prerequisite_failures
)


print(
    "\n=== QUESTION ==="
)

print(
    cell83_question_text
)


print(
    "\n=== RUBRIC ==="
)

print(
    "Version:",
    cell83_rubric.get(
        "version"
    )
)

print(
    "Rubric failures:",
    cell83_rubric_failures
)


print(
    "\n=== HARD GATES ==="
)


for gate_name in (
    cell83_hard_gate_contract.keys()
):

    print(
        gate_name,
        ":",
        (
            "PASS"
            if (
                cell83_hard_gate_results[
                    gate_name
                ]
            )
            else
            "FAIL"
        )
    )


print(
    "Failed hard gates:",
    cell83_failed_hard_gates
)


print(
    "\n=== SEMANTIC GATES ==="
)


for gate_name in (
    cell83_semantic_gate_contract.keys()
):

    judgment = (
        CELL83_SEMANTIC_GATE_JUDGMENTS[
            gate_name
        ]
    )

    print(
        "\n-",
        gate_name,
        ":",
        judgment[
            "status"
        ]
    )

    print(
        "  Reason:",
        judgment[
            "reason"
        ]
    )


print(
    "\nFailed semantic gates:",
    cell83_failed_semantic_gates
)


print(
    "\n=== REQUIRED DIMENSION TESTABILITY ==="
)


for dimension in (
    cell83_required_dimensions
):

    print(
        "-",
        dimension,
        ":",
        CELL83_DIMENSION_TESTABILITY.get(
            dimension
        )
    )


print(
    "Uncovered dimensions:",
    cell83_uncovered_required_dimensions
)


print(
    "\n=== MACHINE DECISION ==="
)

print(
    "Decision:",
    cell83_machine_validation_decision
)

print(
    "Validation record SHA256:",
    cell83_machine_validation_record_sha256
)


print(
    "\n=== DOWNSTREAM BOUNDARY ==="
)

print(
    "Human adjudication:",
    cell83_machine_validation_record[
        "human_adjudication_status"
    ]
)

print(
    "Automatic acceptance:",
    cell83_machine_validation_record[
        "automatic_acceptance"
    ]
)

print(
    "Expected answer:",
    cell83_machine_validation_record[
        "expected_answer_status"
    ]
)

print(
    "Citation validation:",
    cell83_machine_validation_record[
        "citation_target_validation_status"
    ]
)

print(
    "Full 572 generation:",
    cell83_machine_validation_record[
        "full_572_generation"
    ]
)


print(
    "\nSTATUS:",
    cell83_status
)


print(
    "\n=== CODE CELL 83 GATES ==="
)


for gate_name, gate_value in (
    cell83_gates.items()
):

    print(
        gate_name,
        ":",
        (
            "PASS"
            if gate_value
            else "FAIL"
        )
    )


print(
    "\nFAILED GATES:",
    [
        gate_name

        for gate_name, gate_value
        in cell83_gates.items()

        if not gate_value
    ]
)


print(
    "\nCODE CELL 83 GATE:",
    (
        "PASS"
        if cell_83_pass
        else "FAIL"
    )
)


assert cell_83_pass, (
    "Machine Validation failed. Do not perform Human "
    "Adjudication or generate an Expected Answer. Repair this "
    "SAME Code Cell 83 using the printed diagnostics."
)
```

```python
# Code Cell 84

# ============================================================
# Human Adjudication Packet Staging
#
# No automatic human decision.
# No expected-answer generation.
# No citation adjudication.
# No active-state mutation.
# No artifact write.
# ============================================================


# ============================================================
# 1) Prerequisites
# ============================================================

cell84_prerequisite_failures = []


if (
    globals().get(
        "cell_83_pass"
    )
    is not True
):

    cell84_prerequisite_failures.append(
        "CELL83_NOT_PASS"
    )


if (
    globals().get(
        "cell83_status"
    )
    !=
    "MACHINE_VALIDATION_PASS_PENDING_HUMAN_ADJUDICATION"
):

    cell84_prerequisite_failures.append(
        {
            "reason":
                "CELL83_STATUS_UNEXPECTED",

            "actual":
                globals().get(
                    "cell83_status"
                ),
        }
    )


CELL84_REQUIRED_RUNTIME_OBJECTS = [
    "cell82_first_question_draft_record",
    "cell82_question_draft_record_sha256",
    "cell83_machine_validation_record",
    "cell83_machine_validation_record_sha256",
    "cell82_packet",
]


cell84_missing_runtime_objects = [
    name

    for name
    in CELL84_REQUIRED_RUNTIME_OBJECTS

    if (
        name
        not in
        globals()
    )
]


if (
    cell84_missing_runtime_objects
):

    cell84_prerequisite_failures.append(
        {
            "reason":
                "REQUIRED_RUNTIME_OBJECTS_MISSING",

            "objects":
                cell84_missing_runtime_objects,
        }
    )


# ============================================================
# 2) Helper
# ============================================================

def cell84_sha(value):

    return hashlib.sha256(
        stable_json_key(
            value
        ).encode(
            "utf-8"
        )
    ).hexdigest()


# ============================================================
# 3) Freeze state BEFORE review staging
# ============================================================

CELL84_ACTIVE_STATE_OBJECT_NAMES = [
    "golden_case_records",
    "case_groundability_precheck",
    "case_task_design",
    "task_draft_plan_rows",
    "target_evidence_records",
    "pilot_selection_rows",
    "golden_case_records_by_id",
    "task_draft_plans",
    "target_evidence_by_case_id",
    "target_evidence_by_task_draft_id",
]


cell84_active_snapshot_before = (
    cell84_sha(
        {
            name:
                globals()[
                    name
                ]

            for name
            in CELL84_ACTIVE_STATE_OBJECT_NAMES

            if (
                name
                in
                globals()
            )
        }
    )
)


cell84_pilot_snapshot_before = (
    cell84_sha(
        {
            "question_drafting_pilot_packets":
                question_drafting_pilot_packets,

            "question_drafting_pilot_by_id":
                question_drafting_pilot_by_id,
        }
    )
)


cell84_question_record_snapshot_before = (
    cell84_sha(
        cell82_first_question_draft_record
    )
)


# ============================================================
# 4) Resolve Question / Evidence / Machine Validation
# ============================================================

cell84_question_record = (
    cell82_first_question_draft_record
)


cell84_machine_validation = (
    cell83_machine_validation_record
)


cell84_target_parts = (
    cell82_packet.get(
        "target_text_parts",
        [],
    )
)


cell84_review_failures = []


if (
    cell84_machine_validation.get(
        "machine_validation_decision"
    )
    !=
    "PASS"
):

    cell84_review_failures.append(
        {
            "reason":
                "MACHINE_VALIDATION_NOT_PASS",

            "actual":
                cell84_machine_validation.get(
                    "machine_validation_decision"
                ),
        }
    )


if (
    cell84_machine_validation.get(
        "human_adjudication_required"
    )
    is not True
):

    cell84_review_failures.append(
        "HUMAN_ADJUDICATION_NOT_REQUIRED"
    )


if (
    cell84_machine_validation.get(
        "human_adjudication_status"
    )
    !=
    "NOT_STARTED"
):

    cell84_review_failures.append(
        {
            "reason":
                "HUMAN_ADJUDICATION_ALREADY_STARTED",

            "actual":
                cell84_machine_validation.get(
                    "human_adjudication_status"
                ),
        }
    )


if not (
    isinstance(
        cell84_target_parts,
        list,
    )

    and

    len(
        cell84_target_parts
    )
    >
    0
):

    cell84_review_failures.append(
        "TARGET_EVIDENCE_UNAVAILABLE"
    )


# ============================================================
# 5) Human review criteria
# ============================================================

CELL84_HUMAN_REVIEW_CRITERIA = {
    "question_is_natural_and_clear":
        (
            "Question reads naturally and is understandable "
            "without internal pipeline knowledge."
        ),

    "question_matches_intended_task":
        (
            "Question tests the intended retrieval task rather "
            "than unrelated or trivial metadata."
        ),

    "question_is_fully_grounded":
        (
            "All answerable content requested by the question is "
            "present in the validated target evidence."
        ),

    "attribution_is_correct":
        (
            "Named attribution is supported by validated source "
            "metadata and does not promote unsupported authority."
        ),

    "question_does_not_leak_answer":
        (
            "Question does not reveal the substantive answer "
            "components it is supposed to retrieve."
        ),

    "question_has_useful_evaluation_value":
        (
            "Question is suitable for evaluating retrieval hit, "
            "rank, attribution, locator correctness and source "
            "authority correctness."
        ),
}


CELL84_ALLOWED_HUMAN_DECISIONS = [
    "ACCEPT",
    "REJECT",
    "REVISION_REQUIRED",
]


# ============================================================
# 6) Build Human Review Packet
# ============================================================

cell84_human_review_packet = {
    "review_packet_version":
        "question_draft_human_review_packet_v1",

    "question_draft_id":
        cell84_question_record[
            "question_draft_id"
        ],

    "pilot_case_id":
        cell84_question_record[
            "pilot_case_id"
        ],

    "golden_case_id":
        cell84_question_record[
            "golden_case_id"
        ],

    "task_draft_id":
        cell84_question_record[
            "task_draft_id"
        ],

    "target_evidence_id":
        cell84_question_record[
            "target_evidence_id"
        ],

    "question_text":
        cell84_question_record[
            "question_text"
        ],

    "question_text_sha256":
        cell84_question_record[
            "question_text_sha256"
        ],

    "question_draft_record_sha256":
        cell82_question_draft_record_sha256,

    "machine_validation_record_sha256":
        cell83_machine_validation_record_sha256,

    "machine_validation_decision":
        cell84_machine_validation[
            "machine_validation_decision"
        ],

    "hard_gate_results":
        dict(
            cell84_machine_validation[
                "hard_gate_results"
            ]
        ),

    "semantic_gate_results":
        dict(
            cell84_machine_validation[
                "semantic_gate_results"
            ]
        ),

    "required_dimensions":
        list(
            cell82_packet[
                "required_dimensions"
            ]
        ),

    "authority_expectation":
        cell82_packet[
            "authority_expectation"
        ],

    "context_expectation":
        cell82_packet[
            "context_expectation"
        ],

    "evidence_scope":
        cell82_packet[
            "evidence_scope"
        ],

    "target_text_parts":
        [
            {
                "part_index":
                    index,

                "role":
                    part.get(
                        "role"
                    ),

                "text":
                    part.get(
                        "text"
                    ),

                "attribution":
                    part.get(
                        "attribution"
                    ),

                "speaker_state":
                    part.get(
                        "speaker_state"
                    ),

                "locator":
                    part.get(
                        "locator"
                    ),
            }

            for index, part
            in enumerate(
                cell84_target_parts
            )
        ],

    "human_review_criteria":
        dict(
            CELL84_HUMAN_REVIEW_CRITERIA
        ),

    "allowed_human_decisions":
        list(
            CELL84_ALLOWED_HUMAN_DECISIONS
        ),

    "human_decision":
        None,

    "human_decision_reason":
        None,

    "human_adjudication_status":
        "PENDING_HUMAN_DECISION",

    "automatic_acceptance":
        False,

    "expected_answer_status":
        "BLOCKED_PENDING_QUESTION_ACCEPTANCE",

    "citation_target_validation_status":
        "BLOCKED_PENDING_QUESTION_ACCEPTANCE",

    "full_572_generation":
        "BLOCKED",
}


cell84_human_review_packet_sha256 = (
    cell84_sha(
        cell84_human_review_packet
    )
)


# ============================================================
# 7) Validate packet
# ============================================================

cell84_packet_failures = []


if (
    cell84_human_review_packet[
        "question_text_sha256"
    ]
    !=
    hashlib.sha256(
        cell84_human_review_packet[
            "question_text"
        ].encode(
            "utf-8"
        )
    ).hexdigest()
):

    cell84_packet_failures.append(
        "QUESTION_TEXT_HASH_MISMATCH"
    )


if (
    cell84_human_review_packet[
        "machine_validation_decision"
    ]
    !=
    "PASS"
):

    cell84_packet_failures.append(
        "MACHINE_VALIDATION_DECISION_NOT_PASS"
    )


if not all(
    cell84_human_review_packet[
        "hard_gate_results"
    ].values()
):

    cell84_packet_failures.append(
        "NOT_ALL_HARD_GATES_PASS"
    )


if not all(
    cell84_human_review_packet[
        "semantic_gate_results"
    ].values()
):

    cell84_packet_failures.append(
        "NOT_ALL_SEMANTIC_GATES_PASS"
    )


if (
    cell84_human_review_packet[
        "human_decision"
    ]
    is not None
):

    cell84_packet_failures.append(
        "HUMAN_DECISION_WAS_PREMATURELY_ASSIGNED"
    )


if (
    cell84_human_review_packet[
        "automatic_acceptance"
    ]
    is not False
):

    cell84_packet_failures.append(
        "AUTOMATIC_ACCEPTANCE_NOT_FALSE"
    )


# ============================================================
# 8) Freeze state AFTER staging
# ============================================================

cell84_active_snapshot_after = (
    cell84_sha(
        {
            name:
                globals()[
                    name
                ]

            for name
            in CELL84_ACTIVE_STATE_OBJECT_NAMES

            if (
                name
                in
                globals()
            )
        }
    )
)


cell84_pilot_snapshot_after = (
    cell84_sha(
        {
            "question_drafting_pilot_packets":
                question_drafting_pilot_packets,

            "question_drafting_pilot_by_id":
                question_drafting_pilot_by_id,
        }
    )
)


cell84_question_record_snapshot_after = (
    cell84_sha(
        cell82_first_question_draft_record
    )
)


# ============================================================
# 9) Final gates
# ============================================================

cell84_gates = {
    "cell83_is_pass":
        (
            globals().get(
                "cell_83_pass"
            )
            is True
        ),

    "machine_validation_is_pass":
        (
            cell84_machine_validation.get(
                "machine_validation_decision"
            )
            ==
            "PASS"
        ),

    "human_review_is_required":
        (
            cell84_machine_validation.get(
                "human_adjudication_required"
            )
            is True
        ),

    "human_review_packet_is_complete":
        (
            len(
                cell84_packet_failures
            )
            ==
            0
        ),

    "human_decision_is_not_auto_assigned":
        (
            cell84_human_review_packet[
                "human_decision"
            ]
            is None

            and

            cell84_human_review_packet[
                "automatic_acceptance"
            ]
            is False
        ),

    "expected_answer_remains_blocked":
        (
            cell84_human_review_packet[
                "expected_answer_status"
            ]
            ==
            "BLOCKED_PENDING_QUESTION_ACCEPTANCE"
        ),

    "citation_validation_remains_blocked":
        (
            cell84_human_review_packet[
                "citation_target_validation_status"
            ]
            ==
            "BLOCKED_PENDING_QUESTION_ACCEPTANCE"
        ),

    "full_572_generation_remains_blocked":
        (
            cell84_human_review_packet[
                "full_572_generation"
            ]
            ==
            "BLOCKED"
        ),

    "question_record_remains_byte_stable":
        (
            cell84_question_record_snapshot_before
            ==
            cell84_question_record_snapshot_after
        ),

    "active_state_remains_byte_stable":
        (
            cell84_active_snapshot_before
            ==
            cell84_active_snapshot_after
        ),

    "pilot_runtime_remains_byte_stable":
        (
            cell84_pilot_snapshot_before
            ==
            cell84_pilot_snapshot_after
        ),

    "no_artifact_write":
        True,
}


cell_84_pass = all(
    cell84_gates.values()
)


cell84_status = (
    "HUMAN_REVIEW_PACKET_READY_PENDING_EXPLICIT_DECISION"

    if (
        cell_84_pass
    )

    else
    "HUMAN_REVIEW_PACKET_STAGING_FAILED"
)


# ============================================================
# 10) Diagnostics
# ============================================================

print(
    "=== FIRST QUESTION DRAFT — HUMAN REVIEW PACKET ==="
)


print(
    "\n=== PREREQUISITES ==="
)

print(
    "Failures:",
    cell84_prerequisite_failures
)


print(
    "\n=== QUESTION ==="
)

print(
    cell84_human_review_packet[
        "question_text"
    ]
)


print(
    "\n=== TARGET EVIDENCE ==="
)


for part in (
    cell84_human_review_packet[
        "target_text_parts"
    ]
):

    print(
        "\n--- PART",
        part[
            "part_index"
        ],
        "---"
    )

    print(
        part[
            "text"
        ]
    )

    print(
        "Attribution:",
        part[
            "attribution"
        ]
    )


print(
    "\n=== MACHINE VALIDATION ==="
)

print(
    "Decision:",
    cell84_human_review_packet[
        "machine_validation_decision"
    ]
)

print(
    "All hard gates pass:",
    all(
        cell84_human_review_packet[
            "hard_gate_results"
        ].values()
    )
)

print(
    "All semantic gates pass:",
    all(
        cell84_human_review_packet[
            "semantic_gate_results"
        ].values()
    )
)


print(
    "\n=== HUMAN REVIEW CRITERIA ==="
)


for name, description in (
    CELL84_HUMAN_REVIEW_CRITERIA.items()
):

    print(
        "-",
        name,
        ":",
        description
    )


print(
    "\nAllowed human decisions:",
    CELL84_ALLOWED_HUMAN_DECISIONS
)


print(
    "Human decision:",
    cell84_human_review_packet[
        "human_decision"
    ]
)


print(
    "\nReview packet SHA256:",
    cell84_human_review_packet_sha256
)


print(
    "\nSTATUS:",
    cell84_status
)


print(
    "\n=== CODE CELL 84 GATES ==="
)


for gate_name, gate_value in (
    cell84_gates.items()
):

    print(
        gate_name,
        ":",
        (
            "PASS"
            if gate_value
            else "FAIL"
        )
    )


print(
    "\nFAILED GATES:",
    [
        gate_name

        for gate_name, gate_value
        in cell84_gates.items()

        if not gate_value
    ]
)


print(
    "\nCODE CELL 84 GATE:",
    (
        "PASS"
        if cell_84_pass
        else "FAIL"
    )
)


assert cell_84_pass, (
    "Human Review Packet staging failed. "
    "Do not record a Human Decision. "
    "Repair this SAME Code Cell 84 using the printed diagnostics."
)
```

<div style="direction: rtl; white-space: normal; line-height: 1.7;">

<h3>Human Adjudication — ثبت Revision Required و Source-Opacity Rule</h3>

<p>
Human Review نشان داد که Question Draft فعلی از نظر Groundability، Evidence و صحت Attribution درست است، اما با تجربه‌ی کاربری موردنظر محصول سازگار نیست.
</p>

<p>
محصول نهایی یک <strong>مشاور سواد مالی</strong> است. بنابراین User-Facing Question نباید هویت منبع، نویسنده، فایل یا منشأ Evidence را برای کاربر افشا کند.
</p>

<p>
در مقابل، Provenance و Source Authority باید داخل Pipeline به‌طور کامل حفظ شوند تا:
</p>

<ul>
<li>پاسخ فقط از منابع معتبر مورد تأیید تولید شود؛</li>
<li>Groundability قابل بررسی باشد؛</li>
<li>Attribution و Authority به‌صورت داخلی Validate شوند؛</li>
<li>Evaluation بتواند منبع صحیح را بدون نمایش آن به کاربر کنترل کند.</li>
</ul>

<h4>Human Decision</h4>

<p>
Question فعلی به دلیل ذکر مستقیم نام منبع در متن سؤال:
</p>

<p>
<code>REVISION_REQUIRED</code>
</p>

<p>
این Decision به معنی نادرست بودن محتوای سؤال یا Evidence نیست. محتوای factual و Evidence صحیح هستند؛ فقط User-Facing Wording باید Source-Agnostic شود.
</p>

<h4>Rule جدید محصول</h4>

<p>
از این مرحله به بعد، تمام Question Draftهای Golden Dataset باید:
</p>

<ul>
<li>سؤال طبیعی سواد مالی باشند؛</li>
<li>هویت Source / Author را در متن سؤال نمایش ندهند؛</li>
<li>همچنان در Metadata داخلی به Source Authority و Provenance دقیق متصل باشند؛</li>
<li>بدون External Knowledge از Evidence معتبر قابل پاسخ باشند.</li>
</ul>

<p>
این Rule فقط Presentation را از Provenance جدا می‌کند و به معنی حذف Source Attribution از Pipeline نیست.
</p>

</div>

```python
# Code Cell 85

# ============================================================
# Human Adjudication:
# REVISION_REQUIRED because user-facing questions must be
# source-agnostic while internal provenance remains preserved.
#
# No revised question is generated in this cell.
# No expected answer.
# No citation adjudication.
# No active-state mutation.
# No artifact write.
# ============================================================


# ============================================================
# 1) Prerequisites
# ============================================================

cell85_prerequisite_failures = []


if (
    globals().get(
        "cell_84_pass"
    )
    is not True
):

    cell85_prerequisite_failures.append(
        "CELL84_NOT_PASS"
    )


if (
    globals().get(
        "cell84_status"
    )
    !=
    "HUMAN_REVIEW_PACKET_READY_PENDING_EXPLICIT_DECISION"
):

    cell85_prerequisite_failures.append(
        {
            "reason":
                "CELL84_STATUS_UNEXPECTED",

            "actual":
                globals().get(
                    "cell84_status"
                ),
        }
    )


CELL85_REQUIRED_RUNTIME_OBJECTS = [
    "cell84_human_review_packet",
    "cell84_human_review_packet_sha256",
    "cell82_first_question_draft_record",
    "cell83_machine_validation_record",
]


cell85_missing_runtime_objects = [
    name

    for name
    in CELL85_REQUIRED_RUNTIME_OBJECTS

    if (
        name
        not in
        globals()
    )
]


if (
    cell85_missing_runtime_objects
):

    cell85_prerequisite_failures.append(
        {
            "reason":
                "REQUIRED_RUNTIME_OBJECTS_MISSING",

            "objects":
                cell85_missing_runtime_objects,
        }
    )


# ============================================================
# 2) Helper
# ============================================================

def cell85_sha(value):

    return hashlib.sha256(
        stable_json_key(
            value
        ).encode(
            "utf-8"
        )
    ).hexdigest()


# ============================================================
# 3) Freeze state BEFORE adjudication record
# ============================================================

CELL85_ACTIVE_STATE_OBJECT_NAMES = [
    "golden_case_records",
    "case_groundability_precheck",
    "case_task_design",
    "task_draft_plan_rows",
    "target_evidence_records",
    "pilot_selection_rows",
    "golden_case_records_by_id",
    "task_draft_plans",
    "target_evidence_by_case_id",
    "target_evidence_by_task_draft_id",
]


cell85_active_snapshot_before = (
    cell85_sha(
        {
            name:
                globals()[name]

            for name
            in CELL85_ACTIVE_STATE_OBJECT_NAMES

            if (
                name
                in
                globals()
            )
        }
    )
)


cell85_pilot_snapshot_before = (
    cell85_sha(
        {
            "question_drafting_pilot_packets":
                question_drafting_pilot_packets,

            "question_drafting_pilot_by_id":
                question_drafting_pilot_by_id,
        }
    )
)


cell85_question_snapshot_before = (
    cell85_sha(
        cell82_first_question_draft_record
    )
)


# ============================================================
# 4) Explicit Human Decision from user
# ============================================================

CELL85_HUMAN_DECISION = (
    "REVISION_REQUIRED"
)


CELL85_HUMAN_DECISION_REASON = (
    "USER_FACING_FINANCIAL_LITERACY_QUESTION_MUST_NOT_"
    "EXPOSE_SOURCE_OR_AUTHOR_IDENTITY"
)


CELL85_HUMAN_DECISION_DETAIL = (
    "The factual content and validated evidence are correct, "
    "but the question wording directly exposes the source author. "
    "The product is a financial-literacy advisor; source provenance "
    "must remain internal while the user-facing question must be "
    "natural and source-agnostic."
)


# ============================================================
# 5) Freeze product-level Source-Opacity Rule
# ============================================================

QUESTION_USER_FACING_SOURCE_OPACITY_RULE_V1 = {
    "version":
        "question_user_facing_source_opacity_rule_v1",

    "scope":
        "ALL_USER_FACING_GOLDEN_QUESTION_DRAFTS",

    "product_role":
        "FINANCIAL_LITERACY_ADVISOR",

    "user_facing_question_policy":
        "SOURCE_AGNOSTIC",

    "source_identity_disclosure_in_question":
        "FORBIDDEN",

    "author_identity_disclosure_in_question":
        "FORBIDDEN",

    "internal_provenance_retention":
        "REQUIRED",

    "internal_source_authority_validation":
        "REQUIRED",

    "internal_attribution_validation":
        "REQUIRED",

    "evidence_grounding":
        "REQUIRED",

    "external_knowledge":
        "FORBIDDEN",

    "principle":
        (
            "SEPARATE_INTERNAL_PROVENANCE_FROM_"
            "USER_FACING_QUESTION_WORDING"
        ),
}


cell85_source_opacity_rule_sha256 = (
    cell85_sha(
        QUESTION_USER_FACING_SOURCE_OPACITY_RULE_V1
    )
)


# ============================================================
# 6) Validate that current draft violates the new UX rule
# ============================================================

cell85_current_question_text = (
    cell84_human_review_packet[
        "question_text"
    ]
)


cell85_current_target_parts = (
    cell84_human_review_packet[
        "target_text_parts"
    ]
)


cell85_known_source_identity_names = set()


for part in (
    cell85_current_target_parts
):

    attribution = (
        part.get(
            "attribution"
        )
        or
        {}
    )


    for field_name in [
        "author",
        "credited_author_name",
        "knowledge_owner",
        "reference_expert_name",
        "editorial_supervisor_name",
    ]:

        value = (
            attribution.get(
                field_name
            )
        )


        if (
            isinstance(
                value,
                str,
            )

            and

            value.strip()
        ):

            cell85_known_source_identity_names.add(
                value.strip()
            )


cell85_exposed_source_identity_names = [
    name

    for name
    in sorted(
        cell85_known_source_identity_names
    )

    if (
        name
        in
        cell85_current_question_text
    )
]


cell85_revision_basis_failures = []


if (
    len(
        cell85_exposed_source_identity_names
    )
    ==
    0
):

    cell85_revision_basis_failures.append(
        "CURRENT_QUESTION_DOES_NOT_EXPOSE_A_SOURCE_IDENTITY"
    )


if (
    "دکتر کمیل رودی"
    not in
    cell85_current_question_text
):

    cell85_revision_basis_failures.append(
        "EXPECTED_VISIBLE_AUTHOR_NAME_NOT_FOUND_IN_CURRENT_DRAFT"
    )


# ============================================================
# 7) Materialize Human Adjudication Record
# ============================================================

cell85_human_adjudication_record = {
    "adjudication_version":
        "question_draft_human_adjudication_v1",

    "question_draft_id":
        cell84_human_review_packet[
            "question_draft_id"
        ],

    "pilot_case_id":
        cell84_human_review_packet[
            "pilot_case_id"
        ],

    "question_text":
        cell85_current_question_text,

    "question_text_sha256":
        cell84_human_review_packet[
            "question_text_sha256"
        ],

    "human_review_packet_sha256":
        cell84_human_review_packet_sha256,

    "machine_validation_decision":
        cell84_human_review_packet[
            "machine_validation_decision"
        ],

    "human_decision":
        CELL85_HUMAN_DECISION,

    "human_decision_reason":
        CELL85_HUMAN_DECISION_REASON,

    "human_decision_detail":
        CELL85_HUMAN_DECISION_DETAIL,

    "revision_type":
        "USER_FACING_WORDING_ONLY",

    "factual_content_status":
        "VALIDATED_CORRECT",

    "target_evidence_status":
        "RETAIN_UNCHANGED",

    "internal_provenance_status":
        "RETAIN_UNCHANGED",

    "source_opacity_rule_version":
        QUESTION_USER_FACING_SOURCE_OPACITY_RULE_V1[
            "version"
        ],

    "source_opacity_rule_sha256":
        cell85_source_opacity_rule_sha256,

    "exposed_source_identity_names":
        list(
            cell85_exposed_source_identity_names
        ),

    "next_action":
        "CREATE_SOURCE_AGNOSTIC_REVISION",

    "expected_answer_status":
        "BLOCKED_PENDING_REVISED_QUESTION_ACCEPTANCE",

    "citation_target_validation_status":
        "BLOCKED_PENDING_REVISED_QUESTION_ACCEPTANCE",

    "full_572_generation":
        "BLOCKED",
}


cell85_human_adjudication_record_sha256 = (
    cell85_sha(
        cell85_human_adjudication_record
    )
)


# ============================================================
# 8) Validate decision semantics
# ============================================================

cell85_decision_failures = []


if (
    CELL85_HUMAN_DECISION
    !=
    "REVISION_REQUIRED"
):

    cell85_decision_failures.append(
        "HUMAN_DECISION_NOT_REVISION_REQUIRED"
    )


if (
    cell85_human_adjudication_record[
        "factual_content_status"
    ]
    !=
    "VALIDATED_CORRECT"
):

    cell85_decision_failures.append(
        "FACTUAL_CONTENT_WAS_NOT_PRESERVED_AS_CORRECT"
    )


if (
    cell85_human_adjudication_record[
        "target_evidence_status"
    ]
    !=
    "RETAIN_UNCHANGED"
):

    cell85_decision_failures.append(
        "TARGET_EVIDENCE_WAS_NOT_PRESERVED"
    )


if (
    cell85_human_adjudication_record[
        "internal_provenance_status"
    ]
    !=
    "RETAIN_UNCHANGED"
):

    cell85_decision_failures.append(
        "INTERNAL_PROVENANCE_WAS_NOT_PRESERVED"
    )


# ============================================================
# 9) Freeze state AFTER
# ============================================================

cell85_active_snapshot_after = (
    cell85_sha(
        {
            name:
                globals()[name]

            for name
            in CELL85_ACTIVE_STATE_OBJECT_NAMES

            if (
                name
                in
                globals()
            )
        }
    )
)


cell85_pilot_snapshot_after = (
    cell85_sha(
        {
            "question_drafting_pilot_packets":
                question_drafting_pilot_packets,

            "question_drafting_pilot_by_id":
                question_drafting_pilot_by_id,
        }
    )
)


cell85_question_snapshot_after = (
    cell85_sha(
        cell82_first_question_draft_record
    )
)


# ============================================================
# 10) Final gates
# ============================================================

cell85_gates = {
    "cell84_is_pass":
        (
            globals().get(
                "cell_84_pass"
            )
            is True
        ),

    "human_decision_is_revision_required":
        (
            CELL85_HUMAN_DECISION
            ==
            "REVISION_REQUIRED"
        ),

    "revision_basis_is_empirically_present":
        (
            len(
                cell85_revision_basis_failures
            )
            ==
            0
        ),

    "source_opacity_rule_is_frozen":
        (
            QUESTION_USER_FACING_SOURCE_OPACITY_RULE_V1[
                "source_identity_disclosure_in_question"
            ]
            ==
            "FORBIDDEN"

            and

            QUESTION_USER_FACING_SOURCE_OPACITY_RULE_V1[
                "author_identity_disclosure_in_question"
            ]
            ==
            "FORBIDDEN"

            and

            QUESTION_USER_FACING_SOURCE_OPACITY_RULE_V1[
                "internal_provenance_retention"
            ]
            ==
            "REQUIRED"
        ),

    "factual_content_remains_validated":
        (
            cell85_human_adjudication_record[
                "factual_content_status"
            ]
            ==
            "VALIDATED_CORRECT"
        ),

    "target_evidence_is_retained":
        (
            cell85_human_adjudication_record[
                "target_evidence_status"
            ]
            ==
            "RETAIN_UNCHANGED"
        ),

    "internal_provenance_is_retained":
        (
            cell85_human_adjudication_record[
                "internal_provenance_status"
            ]
            ==
            "RETAIN_UNCHANGED"
        ),

    "revised_question_not_generated_yet":
        (
            cell85_human_adjudication_record[
                "next_action"
            ]
            ==
            "CREATE_SOURCE_AGNOSTIC_REVISION"
        ),

    "expected_answer_remains_blocked":
        (
            cell85_human_adjudication_record[
                "expected_answer_status"
            ]
            ==
            "BLOCKED_PENDING_REVISED_QUESTION_ACCEPTANCE"
        ),

    "citation_validation_remains_blocked":
        (
            cell85_human_adjudication_record[
                "citation_target_validation_status"
            ]
            ==
            "BLOCKED_PENDING_REVISED_QUESTION_ACCEPTANCE"
        ),

    "full_572_generation_remains_blocked":
        (
            cell85_human_adjudication_record[
                "full_572_generation"
            ]
            ==
            "BLOCKED"
        ),

    "question_record_remains_byte_stable":
        (
            cell85_question_snapshot_before
            ==
            cell85_question_snapshot_after
        ),

    "active_state_remains_byte_stable":
        (
            cell85_active_snapshot_before
            ==
            cell85_active_snapshot_after
        ),

    "pilot_runtime_remains_byte_stable":
        (
            cell85_pilot_snapshot_before
            ==
            cell85_pilot_snapshot_after
        ),

    "no_artifact_write":
        True,
}


cell_85_pass = all(
    cell85_gates.values()
)


cell85_status = (
    "REVISION_REQUIRED_SOURCE_OPACITY_RULE_FROZEN"

    if (
        cell_85_pass
    )

    else
    "HUMAN_ADJUDICATION_RECORD_FAILED"
)


# ============================================================
# 11) Diagnostics
# ============================================================

print(
    "=== HUMAN ADJUDICATION — SOURCE OPACITY REVISION ==="
)


print(
    "\n=== CURRENT QUESTION ==="
)

print(
    cell85_current_question_text
)


print(
    "\n=== HUMAN DECISION ==="
)

print(
    "Decision:",
    CELL85_HUMAN_DECISION
)

print(
    "Reason:",
    CELL85_HUMAN_DECISION_REASON
)

print(
    "Factual content:",
    cell85_human_adjudication_record[
        "factual_content_status"
    ]
)

print(
    "Target evidence:",
    cell85_human_adjudication_record[
        "target_evidence_status"
    ]
)


print(
    "\n=== SOURCE IDENTITY EXPOSURE ==="
)

print(
    "Known source identities:",
    sorted(
        cell85_known_source_identity_names
    )
)

print(
    "Exposed in question:",
    cell85_exposed_source_identity_names
)

print(
    "Revision-basis failures:",
    cell85_revision_basis_failures
)


print(
    "\n=== SOURCE OPACITY RULE ==="
)

for key, value in (
    QUESTION_USER_FACING_SOURCE_OPACITY_RULE_V1.items()
):

    print(
        key,
        ":",
        value
    )


print(
    "Rule SHA256:",
    cell85_source_opacity_rule_sha256
)


print(
    "\n=== NEXT ACTION ==="
)

print(
    cell85_human_adjudication_record[
        "next_action"
    ]
)

print(
    "Expected answer:",
    cell85_human_adjudication_record[
        "expected_answer_status"
    ]
)

print(
    "Citation validation:",
    cell85_human_adjudication_record[
        "citation_target_validation_status"
    ]
)


print(
    "\nAdjudication record SHA256:",
    cell85_human_adjudication_record_sha256
)


print(
    "\nSTATUS:",
    cell85_status
)


print(
    "\n=== CODE CELL 85 GATES ==="
)


for gate_name, gate_value in (
    cell85_gates.items()
):

    print(
        gate_name,
        ":",
        (
            "PASS"
            if gate_value
            else "FAIL"
        )
    )


print(
    "\nFAILED GATES:",
    [
        gate_name

        for gate_name, gate_value
        in cell85_gates.items()

        if not gate_value
    ]
)


print(
    "\nCODE CELL 85 GATE:",
    (
        "PASS"
        if cell_85_pass
        else "FAIL"
    )
)


assert cell_85_pass, (
    "Human adjudication / source-opacity rule freeze failed. "
    "Do not create a revised Question Draft. Repair this SAME "
    "Code Cell 85 using the printed diagnostics."
)
```

```python
# Code Cell 86

# ============================================================
# Create Source-Agnostic Revision of first Question Draft
#
# Revision only.
# No machine validation yet.
# No human acceptance.
# No expected-answer generation.
# No citation adjudication.
# No active-state mutation.
# No artifact write.
# ============================================================


# ============================================================
# 1) Prerequisites
# ============================================================

cell86_prerequisite_failures = []


if (
    globals().get(
        "cell_85_pass"
    )
    is not True
):

    cell86_prerequisite_failures.append(
        "CELL85_NOT_PASS"
    )


if (
    globals().get(
        "cell85_status"
    )
    !=
    "REVISION_REQUIRED_SOURCE_OPACITY_RULE_FROZEN"
):

    cell86_prerequisite_failures.append(
        {
            "reason":
                "CELL85_STATUS_UNEXPECTED",

            "actual":
                globals().get(
                    "cell85_status"
                ),
        }
    )


CELL86_REQUIRED_RUNTIME_OBJECTS = [
    "cell82_first_question_draft_record",
    "cell82_schema_required_fields",
    "cell82_packet",
    "cell85_human_adjudication_record",
    "cell85_human_adjudication_record_sha256",
    "QUESTION_USER_FACING_SOURCE_OPACITY_RULE_V1",
    "cell85_source_opacity_rule_sha256",
    "cell85_known_source_identity_names",
    "build_question_draft_id",
]


cell86_missing_runtime_objects = [
    name

    for name
    in CELL86_REQUIRED_RUNTIME_OBJECTS

    if (
        name
        not in
        globals()
    )
]


if (
    cell86_missing_runtime_objects
):

    cell86_prerequisite_failures.append(
        {
            "reason":
                "REQUIRED_RUNTIME_OBJECTS_MISSING",

            "objects":
                cell86_missing_runtime_objects,
        }
    )


# ============================================================
# 2) Helper
# ============================================================

def cell86_sha(value):

    return hashlib.sha256(
        stable_json_key(
            value
        ).encode(
            "utf-8"
        )
    ).hexdigest()


# ============================================================
# 3) Freeze state BEFORE revision
# ============================================================

CELL86_ACTIVE_STATE_OBJECT_NAMES = [
    "golden_case_records",
    "case_groundability_precheck",
    "case_task_design",
    "task_draft_plan_rows",
    "target_evidence_records",
    "pilot_selection_rows",
    "golden_case_records_by_id",
    "task_draft_plans",
    "target_evidence_by_case_id",
    "target_evidence_by_task_draft_id",
]


cell86_active_snapshot_before = (
    cell86_sha(
        {
            name:
                globals()[name]

            for name
            in CELL86_ACTIVE_STATE_OBJECT_NAMES

            if (
                name
                in
                globals()
            )
        }
    )
)


cell86_pilot_snapshot_before = (
    cell86_sha(
        {
            "question_drafting_pilot_packets":
                question_drafting_pilot_packets,

            "question_drafting_pilot_by_id":
                question_drafting_pilot_by_id,
        }
    )
)


cell86_original_draft_snapshot_before = (
    cell86_sha(
        cell82_first_question_draft_record
    )
)


# ============================================================
# 4) Validate Human Revision instruction
# ============================================================

cell86_revision_instruction_failures = []


if (
    cell85_human_adjudication_record.get(
        "human_decision"
    )
    !=
    "REVISION_REQUIRED"
):

    cell86_revision_instruction_failures.append(
        "HUMAN_DECISION_NOT_REVISION_REQUIRED"
    )


if (
    cell85_human_adjudication_record.get(
        "revision_type"
    )
    !=
    "USER_FACING_WORDING_ONLY"
):

    cell86_revision_instruction_failures.append(
        {
            "reason":
                "REVISION_TYPE_UNEXPECTED",

            "actual":
                cell85_human_adjudication_record.get(
                    "revision_type"
                ),
        }
    )


if (
    cell85_human_adjudication_record.get(
        "target_evidence_status"
    )
    !=
    "RETAIN_UNCHANGED"
):

    cell86_revision_instruction_failures.append(
        "TARGET_EVIDENCE_NOT_FROZEN"
    )


if (
    cell85_human_adjudication_record.get(
        "internal_provenance_status"
    )
    !=
    "RETAIN_UNCHANGED"
):

    cell86_revision_instruction_failures.append(
        "INTERNAL_PROVENANCE_NOT_FROZEN"
    )


# ============================================================
# 5) Revised user-facing Question
# ============================================================

CELL86_REVISED_QUESTION_TEXT = (
    "برای شروع معاملات در بورس چه مراحلی باید طی شود "
    "و چه روش‌هایی برای خرید و فروش سهام وجود دارد؟"
)


cell86_revised_question_text_sha256 = (
    hashlib.sha256(
        CELL86_REVISED_QUESTION_TEXT.encode(
            "utf-8"
        )
    ).hexdigest()
)


# ============================================================
# 6) Validate Source Opacity
# ============================================================

cell86_source_identity_hits = [
    name

    for name
    in sorted(
        cell85_known_source_identity_names
    )

    if (
        name
        in
        CELL86_REVISED_QUESTION_TEXT
    )
]


CELL86_SOURCE_REFERENCE_CUES = [
    "بر اساس توضیح",
    "بر اساس نظر",
    "به گفته",
    "طبق گفته",
    "طبق نظر",
    "از دیدگاه",
    "از نظر دکتر",
    "از نظر نویسنده",
    "منبع",
    "نویسنده",
]


cell86_source_reference_cue_hits = [
    cue

    for cue
    in CELL86_SOURCE_REFERENCE_CUES

    if (
        cue
        in
        CELL86_REVISED_QUESTION_TEXT
    )
]


cell86_source_opacity_failures = []


if (
    cell86_source_identity_hits
):

    cell86_source_opacity_failures.append(
        {
            "reason":
                "SOURCE_IDENTITY_EXPOSED_IN_REVISED_QUESTION",

            "hits":
                cell86_source_identity_hits,
        }
    )


if (
    cell86_source_reference_cue_hits
):

    cell86_source_opacity_failures.append(
        {
            "reason":
                "SOURCE_REFERENCE_CUE_EXPOSED_IN_REVISED_QUESTION",

            "hits":
                cell86_source_reference_cue_hits,
        }
    )


# ============================================================
# 7) Validate factual scope against unchanged Evidence
# ============================================================

cell86_target_parts = (
    cell82_packet[
        "target_text_parts"
    ]
)


cell86_joined_target_evidence = "\n".join(
    part[
        "text"
    ]

    for part
    in cell86_target_parts
)


CELL86_REQUIRED_EVIDENCE_ANCHORS = [
    "بورس",
    "سامانه سجام",
    "احراز هویت",
    "کد معاملاتی",
    "خرید و فروش سهام",
    "کاغذی",
    "تلفنی",
    "اینترنتی",
]


cell86_missing_evidence_anchors = [
    anchor

    for anchor
    in CELL86_REQUIRED_EVIDENCE_ANCHORS

    if (
        anchor
        not in
        cell86_joined_target_evidence
    )
]


cell86_groundability_failures = []


if (
    cell86_missing_evidence_anchors
):

    cell86_groundability_failures.append(
        {
            "reason":
                "REQUIRED_FACTUAL_SCOPE_NOT_FULLY_PRESENT_IN_TARGET",

            "missing":
                cell86_missing_evidence_anchors,
        }
    )


if not (
    isinstance(
        CELL86_REVISED_QUESTION_TEXT,
        str,
    )

    and

    len(
        CELL86_REVISED_QUESTION_TEXT.strip()
    )
    >=
    20

    and

    CELL86_REVISED_QUESTION_TEXT.endswith(
        "؟"
    )
):

    cell86_groundability_failures.append(
        "REVISED_QUESTION_NOT_NATURAL_NONEMPTY_QUESTION"
    )


# ============================================================
# 8) Ensure wording changed but factual target did not
# ============================================================

cell86_original_question_text = (
    cell82_first_question_draft_record[
        "question_text"
    ]
)


cell86_revision_delta_failures = []


if (
    CELL86_REVISED_QUESTION_TEXT
    ==
    cell86_original_question_text
):

    cell86_revision_delta_failures.append(
        "REVISED_QUESTION_TEXT_DID_NOT_CHANGE"
    )


if (
    cell82_packet[
        "target_evidence_id"
    ]
    !=
    cell85_human_adjudication_record[
        "target_evidence_id"
    ]

    if (
        "target_evidence_id"
        in
        cell85_human_adjudication_record
    )

    else
    False
):

    cell86_revision_delta_failures.append(
        "TARGET_EVIDENCE_ID_CHANGED"
    )


# ============================================================
# 9) Build deterministic revised Question Draft ID
# ============================================================

cell86_revised_question_draft_id = (
    build_question_draft_id(
        pilot_case_id=
            cell82_packet[
                "pilot_case_id"
            ],

        question_text_sha256=
            cell86_revised_question_text_sha256,
    )
)


cell86_revised_question_draft_id_repeat = (
    build_question_draft_id(
        pilot_case_id=
            cell82_packet[
                "pilot_case_id"
            ],

        question_text_sha256=
            cell86_revised_question_text_sha256,
    )
)


cell86_id_failures = []


if not (
    isinstance(
        cell86_revised_question_draft_id,
        str,
    )

    and

    cell86_revised_question_draft_id.startswith(
        "question-draft:"
    )
):

    cell86_id_failures.append(
        {
            "reason":
                "REVISED_QUESTION_DRAFT_ID_INVALID",

            "actual":
                cell86_revised_question_draft_id,
        }
    )


if (
    cell86_revised_question_draft_id
    !=
    cell86_revised_question_draft_id_repeat
):

    cell86_id_failures.append(
        "REVISED_QUESTION_DRAFT_ID_NOT_DETERMINISTIC"
    )


if (
    cell86_revised_question_draft_id
    ==
    cell82_first_question_draft_record[
        "question_draft_id"
    ]
):

    cell86_id_failures.append(
        "REVISED_DRAFT_ID_COLLIDES_WITH_SUPERSEDED_DRAFT"
    )


# ============================================================
# 10) Revised draft initial status
# ============================================================

CELL86_REVISED_INITIAL_STATUSES = {
    "draft_status":
        "DRAFT_CREATED_PENDING_VALIDATION",

    "machine_validation_status":
        "NOT_STARTED",

    "human_adjudication_status":
        "NOT_STARTED",

    "expected_answer_status":
        "BLOCKED_PENDING_QUESTION_ACCEPTANCE",

    "citation_target_validation_status":
        "BLOCKED_PENDING_QUESTION_ACCEPTANCE",
}


# ============================================================
# 11) Build schema field values
# ============================================================

cell86_revised_field_values = {
    "question_draft_id":
        cell86_revised_question_draft_id,

    "question_draft_schema_version":
        "question_draft_record_v1",

    "pilot_case_id":
        cell82_packet[
            "pilot_case_id"
        ],

    "golden_case_id":
        cell82_packet[
            "golden_case_id"
        ],

    "task_draft_id":
        cell82_packet[
            "task_draft_id"
        ],

    "target_evidence_id":
        cell82_packet[
            "target_evidence_id"
        ],

    "source_type":
        cell82_packet[
            "source_type"
        ],

    "stratum":
        cell82_packet[
            "stratum"
        ],

    "task_family":
        cell82_packet[
            "task_family"
        ],

    "question_archetype":
        cell82_packet[
            "question_archetype"
        ],

    "required_dimensions":
        list(
            cell82_packet[
                "required_dimensions"
            ]
        ),

    "authority_expectation":
        cell82_packet[
            "authority_expectation"
        ],

    "context_expectation":
        cell82_packet[
            "context_expectation"
        ],

    "evidence_scope":
        cell82_packet[
            "evidence_scope"
        ],

    "question_text":
        CELL86_REVISED_QUESTION_TEXT,

    "question_text_sha256":
        cell86_revised_question_text_sha256,

    "authoring_mode":
        PILOT_QUESTION_AUTHORING_MODE_V1[
            "mode"
        ],

    "draft_status":
        CELL86_REVISED_INITIAL_STATUSES[
            "draft_status"
        ],

    "machine_validation_status":
        CELL86_REVISED_INITIAL_STATUSES[
            "machine_validation_status"
        ],

    "human_adjudication_status":
        CELL86_REVISED_INITIAL_STATUSES[
            "human_adjudication_status"
        ],

    "expected_answer_status":
        CELL86_REVISED_INITIAL_STATUSES[
            "expected_answer_status"
        ],

    "citation_target_validation_status":
        CELL86_REVISED_INITIAL_STATUSES[
            "citation_target_validation_status"
        ],
}


cell86_schema_failures = []


cell86_unmapped_required_fields = [
    field_name

    for field_name
    in cell82_schema_required_fields

    if (
        field_name
        not in
        cell86_revised_field_values
    )
]


if (
    cell86_unmapped_required_fields
):

    cell86_schema_failures.append(
        {
            "reason":
                "UNMAPPED_REQUIRED_SCHEMA_FIELDS",

            "fields":
                cell86_unmapped_required_fields,
        }
    )


# ============================================================
# 12) Materialize revised Question Draft record
# ============================================================

cell86_revised_question_draft_record = (
    {
        field_name:
            cell86_revised_field_values[
                field_name
            ]

        for field_name
        in cell82_schema_required_fields
    }

    if (
        len(
            cell86_schema_failures
        )
        ==
        0
    )

    else None
)


cell86_revised_question_draft_record_sha256 = (
    cell86_sha(
        cell86_revised_question_draft_record
    )

    if (
        cell86_revised_question_draft_record
        is not None
    )

    else None
)


# ============================================================
# 13) Revision lineage
# ============================================================

cell86_revision_lineage = {
    "revision_lineage_version":
        "question_draft_revision_lineage_v1",

    "revision_number":
        2,

    "supersedes_question_draft_id":
        cell82_first_question_draft_record[
            "question_draft_id"
        ],

    "revised_question_draft_id":
        cell86_revised_question_draft_id,

    "revision_reason":
        CELL85_HUMAN_DECISION_REASON,

    "revision_type":
        "USER_FACING_WORDING_ONLY",

    "human_adjudication_record_sha256":
        cell85_human_adjudication_record_sha256,

    "source_opacity_rule_version":
        QUESTION_USER_FACING_SOURCE_OPACITY_RULE_V1[
            "version"
        ],

    "source_opacity_rule_sha256":
        cell85_source_opacity_rule_sha256,

    "target_evidence_id":
        cell82_packet[
            "target_evidence_id"
        ],

    "target_evidence_change":
        "NONE",

    "internal_provenance_change":
        "NONE",

    "next_action":
        "MACHINE_VALIDATE_REVISED_QUESTION",
}


cell86_revision_lineage_sha256 = (
    cell86_sha(
        cell86_revision_lineage
    )
)


# ============================================================
# 14) Freeze state AFTER revision
# ============================================================

cell86_active_snapshot_after = (
    cell86_sha(
        {
            name:
                globals()[name]

            for name
            in CELL86_ACTIVE_STATE_OBJECT_NAMES

            if (
                name
                in
                globals()
            )
        }
    )
)


cell86_pilot_snapshot_after = (
    cell86_sha(
        {
            "question_drafting_pilot_packets":
                question_drafting_pilot_packets,

            "question_drafting_pilot_by_id":
                question_drafting_pilot_by_id,
        }
    )
)


cell86_original_draft_snapshot_after = (
    cell86_sha(
        cell82_first_question_draft_record
    )
)


# ============================================================
# 15) Final gates
# ============================================================

cell86_gates = {
    "cell85_is_pass":
        (
            globals().get(
                "cell_85_pass"
            )
            is True
        ),

    "human_revision_instruction_is_valid":
        (
            len(
                cell86_revision_instruction_failures
            )
            ==
            0
        ),

    "revised_question_is_source_agnostic":
        (
            len(
                cell86_source_opacity_failures
            )
            ==
            0
        ),

    "revised_question_is_grounded_in_unchanged_target":
        (
            len(
                cell86_groundability_failures
            )
            ==
            0
        ),

    "revision_changes_wording":
        (
            len(
                cell86_revision_delta_failures
            )
            ==
            0

            and

            CELL86_REVISED_QUESTION_TEXT
            !=
            cell86_original_question_text
        ),

    "revised_question_id_is_valid_deterministic_and_distinct":
        (
            len(
                cell86_id_failures
            )
            ==
            0
        ),

    "revised_schema_is_complete":
        (
            len(
                cell86_schema_failures
            )
            ==
            0
        ),

    "revised_question_draft_record_is_materialized":
        (
            cell86_revised_question_draft_record
            is not None
        ),

    "revision_lineage_preserves_same_target":
        (
            cell86_revision_lineage[
                "target_evidence_id"
            ]
            ==
            cell82_packet[
                "target_evidence_id"
            ]

            and

            cell86_revision_lineage[
                "target_evidence_change"
            ]
            ==
            "NONE"

            and

            cell86_revision_lineage[
                "internal_provenance_change"
            ]
            ==
            "NONE"
        ),

    "machine_validation_not_started":
        (
            cell86_revised_question_draft_record[
                "machine_validation_status"
            ]
            ==
            "NOT_STARTED"
        ),

    "human_adjudication_not_started":
        (
            cell86_revised_question_draft_record[
                "human_adjudication_status"
            ]
            ==
            "NOT_STARTED"
        ),

    "expected_answer_remains_blocked":
        (
            cell86_revised_question_draft_record[
                "expected_answer_status"
            ]
            ==
            "BLOCKED_PENDING_QUESTION_ACCEPTANCE"
        ),

    "citation_validation_remains_blocked":
        (
            cell86_revised_question_draft_record[
                "citation_target_validation_status"
            ]
            ==
            "BLOCKED_PENDING_QUESTION_ACCEPTANCE"
        ),

    "superseded_original_draft_remains_byte_stable":
        (
            cell86_original_draft_snapshot_before
            ==
            cell86_original_draft_snapshot_after
        ),

    "active_state_remains_byte_stable":
        (
            cell86_active_snapshot_before
            ==
            cell86_active_snapshot_after
        ),

    "pilot_runtime_remains_byte_stable":
        (
            cell86_pilot_snapshot_before
            ==
            cell86_pilot_snapshot_after
        ),

    "no_artifact_write":
        True,
}


cell_86_pass = all(
    cell86_gates.values()
)


cell86_status = (
    "SOURCE_AGNOSTIC_REVISION_CREATED_PENDING_MACHINE_VALIDATION"

    if (
        cell_86_pass
    )

    else
    "SOURCE_AGNOSTIC_REVISION_CREATION_FAILED"
)


# ============================================================
# 16) Diagnostics
# ============================================================

print(
    "=== SOURCE-AGNOSTIC QUESTION REVISION ==="
)


print(
    "\n=== ORIGINAL QUESTION ==="
)

print(
    cell86_original_question_text
)


print(
    "\n=== REVISED QUESTION ==="
)

print(
    CELL86_REVISED_QUESTION_TEXT
)


print(
    "\n=== SOURCE OPACITY ==="
)

print(
    "Known identities:",
    sorted(
        cell85_known_source_identity_names
    )
)

print(
    "Identity hits:",
    cell86_source_identity_hits
)

print(
    "Source-reference cue hits:",
    cell86_source_reference_cue_hits
)

print(
    "Failures:",
    cell86_source_opacity_failures
)


print(
    "\n=== GROUNDABILITY ==="
)

print(
    "Missing evidence anchors:",
    cell86_missing_evidence_anchors
)

print(
    "Failures:",
    cell86_groundability_failures
)


print(
    "\n=== DRAFT IDS ==="
)

print(
    "Superseded Draft ID:",
    cell82_first_question_draft_record[
        "question_draft_id"
    ]
)

print(
    "Revised Draft ID:",
    cell86_revised_question_draft_id
)

print(
    "Repeat ID:",
    cell86_revised_question_draft_id_repeat
)

print(
    "ID failures:",
    cell86_id_failures
)


print(
    "\n=== REVISED RECORD ==="
)

print(
    "Schema fields:",
    (
        list(
            cell86_revised_question_draft_record.keys()
        )

        if (
            cell86_revised_question_draft_record
            is not None
        )

        else []
    )
)

print(
    "Record SHA256:",
    cell86_revised_question_draft_record_sha256
)


print(
    "\n=== REVISION LINEAGE ==="
)

for key, value in (
    cell86_revision_lineage.items()
):

    print(
        key,
        ":",
        value
    )


print(
    "Revision lineage SHA256:",
    cell86_revision_lineage_sha256
)


print(
    "\n=== DOWNSTREAM BOUNDARY ==="
)

print(
    "Machine validation:",
    cell86_revised_question_draft_record[
        "machine_validation_status"
    ]
)

print(
    "Human adjudication:",
    cell86_revised_question_draft_record[
        "human_adjudication_status"
    ]
)

print(
    "Expected answer:",
    cell86_revised_question_draft_record[
        "expected_answer_status"
    ]
)

print(
    "Citation validation:",
    cell86_revised_question_draft_record[
        "citation_target_validation_status"
    ]
)


print(
    "\nSTATUS:",
    cell86_status
)


print(
    "\n=== CODE CELL 86 GATES ==="
)


for gate_name, gate_value in (
    cell86_gates.items()
):

    print(
        gate_name,
        ":",
        (
            "PASS"
            if gate_value
            else "FAIL"
        )
    )


print(
    "\nFAILED GATES:",
    [
        gate_name

        for gate_name, gate_value
        in cell86_gates.items()

        if not gate_value
    ]
)


print(
    "\nCODE CELL 86 GATE:",
    (
        "PASS"
        if cell_86_pass
        else "FAIL"
    )
)


assert cell_86_pass, (
    "Source-Agnostic Question Revision failed. "
    "Do not Machine Validate the revised draft. "
    "Repair this SAME Code Cell 86 using the printed diagnostics."
)
```

<div style="direction: rtl; white-space: normal; line-height: 1.7;">

<h3>Machine Validation مجدد برای Source-Agnostic Revision</h3>

<p>
Cell 86 نسخه‌ی اصلاح‌شده‌ی اولین Question Draft را ساخت و نام نویسنده و هرگونه اشاره‌ی مستقیم به Source را از User-Facing Question حذف کرد، در حالی که Target Evidence، Attribution و Provenance داخلی بدون تغییر باقی ماندند.
</p>

<p>
در این مرحله Revision جدید دوباره Machine Validate می‌شود.
</p>

<h4>اصل مهم این Validation</h4>

<p>
از اینجا به بعد دو لایه باید کاملاً از هم جدا باشند:
</p>

<ul>
<li><strong>User-Facing Layer:</strong> سؤال باید طبیعی، مالی و Source-Agnostic باشد.</li>
<li><strong>Internal Evaluation Layer:</strong> Source Authority، Attribution، Locator و Provenance همچنان باید دقیق و قابل ارزیابی باشند.</li>
</ul>

<p>
بنابراین Dimensionهایی مانند
<code>attribution_correctness</code>
و
<code>source_authority_correctness</code>
دیگر به معنی نمایش نام منبع در سؤال نیستند؛ این Dimensionها به‌صورت داخلی بررسی می‌کنند که Retrieval و Evidence واقعاً از Source معتبر مورد انتظار آمده باشند.
</p>

<h4>Validation مورد انتظار</h4>

<ul>
<li>تمام Hard Gateهای Question Drafting باید PASS شوند؛</li>
<li>تمام Semantic Gateها باید PASS شوند؛</li>
<li>Source Identity در Question Text نباید دیده شود؛</li>
<li>Internal Provenance باید کامل و معتبر باقی بماند؛</li>
<li>Question باید فقط از Target Evidence قابل پاسخ باشد؛</li>
<li>Target Evidence نباید نسبت به Revision قبلی تغییر کرده باشد؛</li>
<li>Expected Answer و Citation Validation همچنان Blocked باقی بمانند.</li>
</ul>

<p>
اگر این Cell PASS شود، Revision جدید آماده‌ی Human Acceptance نهایی خواهد بود.
</p>

</div>

```python
# Code Cell 87

# ============================================================
# Machine Validation of Source-Agnostic Revision
#
# Source opacity is now an explicit product-level gate.
#
# Internal provenance / authority remain required.
# No human acceptance.
# No expected-answer generation.
# No citation adjudication.
# No active-state mutation.
# No artifact write.
# ============================================================


# ============================================================
# 1) Prerequisites
# ============================================================

cell87_prerequisite_failures = []


if (
    globals().get(
        "cell_86_pass"
    )
    is not True
):

    cell87_prerequisite_failures.append(
        "CELL86_NOT_PASS"
    )


if (
    globals().get(
        "cell86_status"
    )
    !=
    "SOURCE_AGNOSTIC_REVISION_CREATED_PENDING_MACHINE_VALIDATION"
):

    cell87_prerequisite_failures.append(
        {
            "reason":
                "CELL86_STATUS_UNEXPECTED",

            "actual":
                globals().get(
                    "cell86_status"
                ),
        }
    )


CELL87_REQUIRED_RUNTIME_OBJECTS = [
    "cell86_revised_question_draft_record",
    "cell86_revised_question_draft_record_sha256",
    "cell86_revision_lineage",
    "cell86_revision_lineage_sha256",
    "cell82_packet",
    "cell81_selected_task",
    "QUESTION_DRAFT_ACCEPTANCE_RUBRIC_V1",
    "QUESTION_USER_FACING_SOURCE_OPACITY_RULE_V1",
    "cell85_source_opacity_rule_sha256",
    "cell85_known_source_identity_names",
]


cell87_missing_runtime_objects = [
    name

    for name
    in CELL87_REQUIRED_RUNTIME_OBJECTS

    if (
        name
        not in
        globals()
    )
]


if (
    cell87_missing_runtime_objects
):

    cell87_prerequisite_failures.append(
        {
            "reason":
                "REQUIRED_RUNTIME_OBJECTS_MISSING",

            "objects":
                cell87_missing_runtime_objects,
        }
    )


# ============================================================
# 2) Helper
# ============================================================

def cell87_sha(value):

    return hashlib.sha256(
        stable_json_key(
            value
        ).encode(
            "utf-8"
        )
    ).hexdigest()


# ============================================================
# 3) Freeze state BEFORE validation
# ============================================================

CELL87_ACTIVE_STATE_OBJECT_NAMES = [
    "golden_case_records",
    "case_groundability_precheck",
    "case_task_design",
    "task_draft_plan_rows",
    "target_evidence_records",
    "pilot_selection_rows",
    "golden_case_records_by_id",
    "task_draft_plans",
    "target_evidence_by_case_id",
    "target_evidence_by_task_draft_id",
]


cell87_active_snapshot_before = (
    cell87_sha(
        {
            name:
                globals()[name]

            for name
            in CELL87_ACTIVE_STATE_OBJECT_NAMES

            if (
                name
                in
                globals()
            )
        }
    )
)


cell87_pilot_snapshot_before = (
    cell87_sha(
        {
            "question_drafting_pilot_packets":
                question_drafting_pilot_packets,

            "question_drafting_pilot_by_id":
                question_drafting_pilot_by_id,
        }
    )
)


cell87_revised_draft_snapshot_before = (
    cell87_sha(
        cell86_revised_question_draft_record
    )
)


# ============================================================
# 4) Resolve revised Question / Target / Rubric
# ============================================================

cell87_question_record = (
    cell86_revised_question_draft_record
)


cell87_question_text = (
    cell87_question_record[
        "question_text"
    ]
)


cell87_target_parts = (
    cell82_packet[
        "target_text_parts"
    ]
)


cell87_joined_evidence = "\n".join(
    part[
        "text"
    ]

    for part
    in cell87_target_parts
)


cell87_rubric = (
    QUESTION_DRAFT_ACCEPTANCE_RUBRIC_V1
)


cell87_hard_gate_contract = (
    cell87_rubric.get(
        "hard_gates",
        {},
    )
)


cell87_semantic_gate_contract = (
    cell87_rubric.get(
        "semantic_gates",
        {},
    )
)


# ============================================================
# 5) Validate internal provenance
# ============================================================

cell87_internal_attribution_names = set()

cell87_valid_direct_author_parts = []


for part in (
    cell87_target_parts
):

    attribution = (
        part.get(
            "attribution"
        )
        or
        {}
    )


    for field_name in [
        "author",
        "credited_author_name",
        "knowledge_owner",
    ]:

        value = (
            attribution.get(
                field_name
            )
        )


        if (
            isinstance(
                value,
                str,
            )

            and

            value.strip()
        ):

            cell87_internal_attribution_names.add(
                value.strip()
            )


    if (
        attribution.get(
            "attribution_status"
        )
        ==
        "policy_validated"

        and

        attribution.get(
            "content_authority_type"
        )
        ==
        "direct_author"
    ):

        cell87_valid_direct_author_parts.append(
            part
        )


cell87_internal_provenance_failures = []


if (
    len(
        cell87_valid_direct_author_parts
    )
    ==
    0
):

    cell87_internal_provenance_failures.append(
        "NO_POLICY_VALIDATED_DIRECT_AUTHOR_EVIDENCE"
    )


if (
    cell82_packet.get(
        "authority_expectation"
    )
    !=
    "PRIMARY_DIRECT_AUTHOR"
):

    cell87_internal_provenance_failures.append(
        {
            "reason":
                "AUTHORITY_EXPECTATION_UNEXPECTED",

            "actual":
                cell82_packet.get(
                    "authority_expectation"
                ),
        }
    )


# ============================================================
# 6) Source-opacity product gates
# ============================================================

CELL87_SOURCE_REFERENCE_CUES = [
    "بر اساس توضیح",
    "بر اساس نظر",
    "به گفته",
    "طبق گفته",
    "طبق نظر",
    "از دیدگاه",
    "از نظر دکتر",
    "از نظر نویسنده",
    "منبع",
    "نویسنده",
]


cell87_source_identity_hits = [
    name

    for name
    in sorted(
        cell85_known_source_identity_names
    )

    if (
        name
        in
        cell87_question_text
    )
]


cell87_source_reference_cue_hits = [
    cue

    for cue
    in CELL87_SOURCE_REFERENCE_CUES

    if (
        cue
        in
        cell87_question_text
    )
]


cell87_source_opacity_gate_results = {
    "source_identity_not_exposed":
        (
            len(
                cell87_source_identity_hits
            )
            ==
            0
        ),

    "source_reference_cue_not_exposed":
        (
            len(
                cell87_source_reference_cue_hits
            )
            ==
            0
        ),

    "user_facing_question_policy_is_source_agnostic":
        (
            QUESTION_USER_FACING_SOURCE_OPACITY_RULE_V1.get(
                "user_facing_question_policy"
            )
            ==
            "SOURCE_AGNOSTIC"
        ),

    "internal_provenance_is_required":
        (
            QUESTION_USER_FACING_SOURCE_OPACITY_RULE_V1.get(
                "internal_provenance_retention"
            )
            ==
            "REQUIRED"
        ),

    "internal_authority_validation_is_required":
        (
            QUESTION_USER_FACING_SOURCE_OPACITY_RULE_V1.get(
                "internal_source_authority_validation"
            )
            ==
            "REQUIRED"
        ),

    "internal_attribution_validation_is_required":
        (
            QUESTION_USER_FACING_SOURCE_OPACITY_RULE_V1.get(
                "internal_attribution_validation"
            )
            ==
            "REQUIRED"
        ),
}


cell87_failed_source_opacity_gates = [
    gate_name

    for gate_name, gate_value
    in cell87_source_opacity_gate_results.items()

    if not gate_value
]


# ============================================================
# 7) Evidence anchors
# ============================================================

CELL87_REQUIRED_ANSWER_EVIDENCE_ANCHORS = [
    "سامانه سجام",
    "احراز هویت",
    "کد معاملاتی",
    "کاغذی",
    "تلفنی",
    "اینترنتی",
]


cell87_missing_evidence_anchors = [
    anchor

    for anchor
    in CELL87_REQUIRED_ANSWER_EVIDENCE_ANCHORS

    if (
        anchor
        not in
        cell87_joined_evidence
    )
]


CELL87_ANSWER_LEAKAGE_ANCHORS = [
    "سامانه سجام",
    "احراز هویت",
    "کد معاملاتی",
    "روش کاغذی",
    "روش تلفنی",
    "روش اینترنتی",
]


cell87_answer_leakage_hits = [
    anchor

    for anchor
    in CELL87_ANSWER_LEAKAGE_ANCHORS

    if (
        anchor
        in
        cell87_question_text
    )
]


CELL87_INTERNAL_METADATA_MARKERS = [
    "golden-case:",
    "task-draft:",
    "target-evidence:",
    "question-pilot:",
    "question-draft:",
    "sha256",
    "candidate_identity",
    "document_id",
    "source_id",
]


cell87_internal_metadata_hits = [
    marker

    for marker
    in CELL87_INTERNAL_METADATA_MARKERS

    if (
        marker
        in
        cell87_question_text
    )
]


CELL87_LOCATOR_MARKERS = [
    "http://",
    "https://",
    "www.",
    "صفحه ",
    "خط ",
    "فایل ",
]


cell87_locator_hits = [
    marker

    for marker
    in CELL87_LOCATOR_MARKERS

    if (
        marker
        in
        cell87_question_text
    )
]


# ============================================================
# 8) Hard Gates
# ============================================================

cell87_hard_gate_results = {
    "nonempty_question_text":
        (
            isinstance(
                cell87_question_text,
                str,
            )

            and

            len(
                cell87_question_text.strip()
            )
            >
            0
        ),

    "natural_user_question":
        (
            isinstance(
                cell87_question_text,
                str,
            )

            and

            len(
                cell87_question_text.strip()
            )
            >=
            20

            and

            cell87_question_text.strip().endswith(
                "؟"
            )

            and

            "\n"
            not in
            cell87_question_text.strip()
        ),

    "external_knowledge_dependency":
        (
            len(
                cell87_missing_evidence_anchors
            )
            ==
            0
        ),

    "answer_leakage_in_question":
        (
            len(
                cell87_answer_leakage_hits
            )
            ==
            0
        ),

    "internal_document_id_leakage":
        (
            not any(
                marker
                in
                cell87_question_text

                for marker
                in [
                    "document_id",
                    "source_id",
                    "golden-case:",
                    "task-draft:",
                    "target-evidence:",
                    "question-pilot:",
                ]
            )
        ),

    "internal_pipeline_metadata_leakage":
        (
            len(
                cell87_internal_metadata_hits
            )
            ==
            0
        ),

    "unknown_speaker_inference":
        (
            "گوینده"
            not in
            cell87_question_text

            and

            "speaker"
            not in
            cell87_question_text.lower()
        ),

    # Attribution is validated internally.
    # It is intentionally NOT exposed in the user-facing question.
    "unsupported_attribution":
        (
            len(
                cell87_internal_attribution_names
            )
            >
            0

            and

            len(
                cell87_valid_direct_author_parts
            )
            >
            0
        ),

    "unsupported_authority_promotion":
        (
            cell82_packet.get(
                "authority_expectation"
            )
            ==
            "PRIMARY_DIRECT_AUTHOR"

            and

            len(
                cell87_valid_direct_author_parts
            )
            >=
            1
        ),

    "locator_fabrication":
        (
            len(
                cell87_locator_hits
            )
            ==
            0
        ),

    "question_must_be_groundable_in_target_text_parts":
        (
            len(
                cell87_target_parts
            )
            >=
            1

            and

            len(
                cell87_missing_evidence_anchors
            )
            ==
            0
        ),

    "target_evidence_must_already_be_validated":
        (
            cell81_selected_task.get(
                "target_evidence_status"
            )
            ==
            "SELECTED_AND_VALIDATED"
        ),
}


cell87_failed_hard_gates = [
    gate_name

    for gate_name, gate_value
    in cell87_hard_gate_results.items()

    if not gate_value
]


# ============================================================
# 9) Required dimension testability
#
# Important:
# Attribution / authority / locator dimensions are INTERNAL
# evaluation dimensions and do not require source exposure in
# the user-facing Question.
# ============================================================

cell87_required_dimensions = (
    cell82_packet[
        "required_dimensions"
    ]
)


CELL87_DIMENSION_TESTABILITY = {
    "retrieval_hit":
        (
            "The question requires explicit factual evidence "
            "from the validated target document."
        ),

    "rank":
        (
            "The validated target document can be checked for "
            "ranking against competing retrieval candidates."
        ),

    "attribution_correctness":
        (
            "Retrieved evidence attribution can be checked "
            "internally against policy-validated provenance "
            "without exposing the author to the user."
        ),

    "citation_locator_correctness":
        (
            "Retrieved evidence can be checked internally against "
            "the exact validated target locator."
        ),

    "source_authority_correctness":
        (
            "Retrieved evidence authority can be checked internally "
            "against PRIMARY_DIRECT_AUTHOR without source disclosure."
        ),
}


cell87_uncovered_required_dimensions = [
    dimension

    for dimension
    in cell87_required_dimensions

    if (
        dimension
        not in
        CELL87_DIMENSION_TESTABILITY
    )
]


# ============================================================
# 10) Semantic Gates
# ============================================================

CELL87_SEMANTIC_GATE_JUDGMENTS = {
    "task_family_alignment": {
        "status":
            "PASS",

        "reason":
            (
                "The revised question remains a direct factual "
                "retrieval question over the same validated target."
            ),
    },

    "question_archetype_alignment": {
        "status":
            "PASS",

        "reason":
            (
                "The authority constraint is enforced internally "
                "through provenance and target selection while the "
                "user-facing wording remains source-agnostic."
            ),
    },

    "evidence_scope_alignment": {
        "status":
            "PASS",

        "reason":
            (
                "All information required to answer the question "
                "is contained in the same single source document."
            ),
    },

    "authority_expectation_alignment": {
        "status":
            (
                "PASS"

                if (
                    len(
                        cell87_internal_provenance_failures
                    )
                    ==
                    0
                )

                else
                "FAIL"
            ),

        "reason":
            (
                "PRIMARY_DIRECT_AUTHOR is preserved and validated "
                "internally without exposing the author to the user."
            ),
    },

    "context_expectation_alignment": {
        "status":
            "PASS",

        "reason":
            (
                "The question requires only the minimal direct "
                "evidence contained in the selected target."
            ),
    },

    "required_dimensions_are_testable_by_question": {
        "status":
            (
                "PASS"

                if (
                    len(
                        cell87_uncovered_required_dimensions
                    )
                    ==
                    0
                )

                else
                "FAIL"
            ),

        "reason":
            (
                "All required dimensions remain testable; provenance, "
                "authority and locator dimensions are evaluated "
                "internally rather than through source disclosure."
            ),
    },

    "question_is_not_trivial_metadata_repetition": {
        "status":
            "PASS",

        "reason":
            (
                "The question requests substantive financial-literacy "
                "content and contains no pipeline metadata."
            ),
    },

    "question_is_answerable_without_external_context": {
        "status":
            (
                "PASS"

                if (
                    len(
                        cell87_missing_evidence_anchors
                    )
                    ==
                    0
                )

                else
                "FAIL"
            ),

        "reason":
            (
                "The validated target evidence contains every "
                "required answer component."
            ),
    },
}


cell87_semantic_gate_results = {
    gate_name:
        (
            judgment[
                "status"
            ]
            ==
            "PASS"
        )

    for gate_name, judgment
    in CELL87_SEMANTIC_GATE_JUDGMENTS.items()
}


cell87_failed_semantic_gates = [
    gate_name

    for gate_name, gate_value
    in cell87_semantic_gate_results.items()

    if not gate_value
]


# ============================================================
# 11) Contract coverage
# ============================================================

cell87_contract_failures = []


if (
    set(
        cell87_hard_gate_results.keys()
    )
    !=
    set(
        cell87_hard_gate_contract.keys()
    )
):

    cell87_contract_failures.append(
        {
            "reason":
                "HARD_GATE_COVERAGE_NOT_EXACT",

            "expected":
                sorted(
                    cell87_hard_gate_contract.keys()
                ),

            "actual":
                sorted(
                    cell87_hard_gate_results.keys()
                ),
        }
    )


if (
    set(
        cell87_semantic_gate_results.keys()
    )
    !=
    set(
        cell87_semantic_gate_contract.keys()
    )
):

    cell87_contract_failures.append(
        {
            "reason":
                "SEMANTIC_GATE_COVERAGE_NOT_EXACT",

            "expected":
                sorted(
                    cell87_semantic_gate_contract.keys()
                ),

            "actual":
                sorted(
                    cell87_semantic_gate_results.keys()
                ),
        }
    )


# ============================================================
# 12) Machine decision
# ============================================================

cell87_machine_validation_decision = (
    "PASS"

    if (
        len(
            cell87_prerequisite_failures
        )
        ==
        0

        and

        len(
            cell87_internal_provenance_failures
        )
        ==
        0

        and

        len(
            cell87_failed_source_opacity_gates
        )
        ==
        0

        and

        len(
            cell87_failed_hard_gates
        )
        ==
        0

        and

        len(
            cell87_failed_semantic_gates
        )
        ==
        0

        and

        len(
            cell87_uncovered_required_dimensions
        )
        ==
        0

        and

        len(
            cell87_contract_failures
        )
        ==
        0
    )

    else
    "FAIL"
)


# ============================================================
# 13) Materialize revised Machine Validation record
# ============================================================

cell87_machine_validation_record = {
    "validation_version":
        "question_draft_machine_validation_v2_source_opaque",

    "question_draft_id":
        cell86_revised_question_draft_record[
            "question_draft_id"
        ],

    "pilot_case_id":
        cell86_revised_question_draft_record[
            "pilot_case_id"
        ],

    "question_text":
        cell87_question_text,

    "question_text_sha256":
        cell86_revised_question_draft_record[
            "question_text_sha256"
        ],

    "question_draft_record_sha256":
        cell86_revised_question_draft_record_sha256,

    "revision_lineage_sha256":
        cell86_revision_lineage_sha256,

    "rubric_version":
        cell87_rubric[
            "version"
        ],

    "source_opacity_rule_version":
        QUESTION_USER_FACING_SOURCE_OPACITY_RULE_V1[
            "version"
        ],

    "source_opacity_rule_sha256":
        cell85_source_opacity_rule_sha256,

    "source_opacity_gate_results":
        dict(
            cell87_source_opacity_gate_results
        ),

    "hard_gate_results":
        dict(
            cell87_hard_gate_results
        ),

    "semantic_gate_results":
        dict(
            cell87_semantic_gate_results
        ),

    "semantic_gate_judgments":
        dict(
            CELL87_SEMANTIC_GATE_JUDGMENTS
        ),

    "required_dimension_testability":
        {
            dimension:
                CELL87_DIMENSION_TESTABILITY[
                    dimension
                ]

            for dimension
            in cell87_required_dimensions
        },

    "internal_provenance_status":
        (
            "VALIDATED_AND_HIDDEN_FROM_USER"
        ),

    "machine_validation_decision":
        cell87_machine_validation_decision,

    "human_adjudication_required":
        True,

    "human_adjudication_status":
        "NOT_STARTED",

    "automatic_acceptance":
        False,

    "expected_answer_status":
        "BLOCKED_PENDING_QUESTION_ACCEPTANCE",

    "citation_target_validation_status":
        "BLOCKED_PENDING_QUESTION_ACCEPTANCE",

    "full_572_generation":
        "BLOCKED",
}


cell87_machine_validation_record_sha256 = (
    cell87_sha(
        cell87_machine_validation_record
    )
)


# ============================================================
# 14) Freeze state AFTER validation
# ============================================================

cell87_active_snapshot_after = (
    cell87_sha(
        {
            name:
                globals()[name]

            for name
            in CELL87_ACTIVE_STATE_OBJECT_NAMES

            if (
                name
                in
                globals()
            )
        }
    )
)


cell87_pilot_snapshot_after = (
    cell87_sha(
        {
            "question_drafting_pilot_packets":
                question_drafting_pilot_packets,

            "question_drafting_pilot_by_id":
                question_drafting_pilot_by_id,
        }
    )
)


cell87_revised_draft_snapshot_after = (
    cell87_sha(
        cell86_revised_question_draft_record
    )
)


# ============================================================
# 15) Final gates
# ============================================================

cell87_gates = {
    "cell86_is_pass":
        (
            globals().get(
                "cell_86_pass"
            )
            is True
        ),

    "source_opacity_rule_is_enforced":
        (
            len(
                cell87_failed_source_opacity_gates
            )
            ==
            0
        ),

    "internal_provenance_remains_valid":
        (
            len(
                cell87_internal_provenance_failures
            )
            ==
            0
        ),

    "all_frozen_hard_gates_pass":
        (
            len(
                cell87_failed_hard_gates
            )
            ==
            0
        ),

    "all_frozen_semantic_gates_pass":
        (
            len(
                cell87_failed_semantic_gates
            )
            ==
            0
        ),

    "all_required_dimensions_remain_testable":
        (
            len(
                cell87_uncovered_required_dimensions
            )
            ==
            0
        ),

    "rubric_contract_coverage_is_exact":
        (
            len(
                cell87_contract_failures
            )
            ==
            0
        ),

    "machine_validation_decision_is_pass":
        (
            cell87_machine_validation_decision
            ==
            "PASS"
        ),

    "human_acceptance_is_still_required":
        (
            cell87_machine_validation_record[
                "human_adjudication_required"
            ]
            is True

            and

            cell87_machine_validation_record[
                "human_adjudication_status"
            ]
            ==
            "NOT_STARTED"

            and

            cell87_machine_validation_record[
                "automatic_acceptance"
            ]
            is False
        ),

    "expected_answer_remains_blocked":
        (
            cell87_machine_validation_record[
                "expected_answer_status"
            ]
            ==
            "BLOCKED_PENDING_QUESTION_ACCEPTANCE"
        ),

    "citation_validation_remains_blocked":
        (
            cell87_machine_validation_record[
                "citation_target_validation_status"
            ]
            ==
            "BLOCKED_PENDING_QUESTION_ACCEPTANCE"
        ),

    "full_572_generation_remains_blocked":
        (
            cell87_machine_validation_record[
                "full_572_generation"
            ]
            ==
            "BLOCKED"
        ),

    "revised_question_record_remains_byte_stable":
        (
            cell87_revised_draft_snapshot_before
            ==
            cell87_revised_draft_snapshot_after
        ),

    "active_state_remains_byte_stable":
        (
            cell87_active_snapshot_before
            ==
            cell87_active_snapshot_after
        ),

    "pilot_runtime_remains_byte_stable":
        (
            cell87_pilot_snapshot_before
            ==
            cell87_pilot_snapshot_after
        ),

    "no_artifact_write":
        True,
}


cell_87_pass = all(
    cell87_gates.values()
)


cell87_status = (
    "SOURCE_AGNOSTIC_REVISION_MACHINE_PASS_PENDING_HUMAN_ACCEPTANCE"

    if (
        cell_87_pass
    )

    else
    "SOURCE_AGNOSTIC_REVISION_MACHINE_VALIDATION_FAILED"
)


# ============================================================
# 16) Diagnostics
# ============================================================

print(
    "=== SOURCE-AGNOSTIC REVISION — MACHINE VALIDATION ==="
)


print(
    "\n=== QUESTION ==="
)

print(
    cell87_question_text
)


print(
    "\n=== SOURCE OPACITY GATES ==="
)


for gate_name, gate_value in (
    cell87_source_opacity_gate_results.items()
):

    print(
        gate_name,
        ":",
        (
            "PASS"
            if gate_value
            else "FAIL"
        )
    )


print(
    "Identity hits:",
    cell87_source_identity_hits
)

print(
    "Source-reference cue hits:",
    cell87_source_reference_cue_hits
)

print(
    "Failed source-opacity gates:",
    cell87_failed_source_opacity_gates
)


print(
    "\n=== INTERNAL PROVENANCE ==="
)

print(
    "Attribution names:",
    sorted(
        cell87_internal_attribution_names
    )
)

print(
    "Validated direct-author parts:",
    len(
        cell87_valid_direct_author_parts
    )
)

print(
    "Failures:",
    cell87_internal_provenance_failures
)


print(
    "\n=== HARD GATES ==="
)


for gate_name in (
    cell87_hard_gate_contract.keys()
):

    print(
        gate_name,
        ":",
        (
            "PASS"
            if (
                cell87_hard_gate_results[
                    gate_name
                ]
            )
            else
            "FAIL"
        )
    )


print(
    "Failed hard gates:",
    cell87_failed_hard_gates
)


print(
    "\n=== SEMANTIC GATES ==="
)


for gate_name in (
    cell87_semantic_gate_contract.keys()
):

    judgment = (
        CELL87_SEMANTIC_GATE_JUDGMENTS[
            gate_name
        ]
    )


    print(
        "\n-",
        gate_name,
        ":",
        judgment[
            "status"
        ]
    )

    print(
        "  Reason:",
        judgment[
            "reason"
        ]
    )


print(
    "\nFailed semantic gates:",
    cell87_failed_semantic_gates
)


print(
    "\n=== REQUIRED DIMENSION TESTABILITY ==="
)


for dimension in (
    cell87_required_dimensions
):

    print(
        "-",
        dimension,
        ":",
        CELL87_DIMENSION_TESTABILITY.get(
            dimension
        )
    )


print(
    "Uncovered dimensions:",
    cell87_uncovered_required_dimensions
)


print(
    "\n=== MACHINE DECISION ==="
)

print(
    "Decision:",
    cell87_machine_validation_decision
)

print(
    "Validation record SHA256:",
    cell87_machine_validation_record_sha256
)


print(
    "\n=== DOWNSTREAM BOUNDARY ==="
)

print(
    "Human adjudication:",
    cell87_machine_validation_record[
        "human_adjudication_status"
    ]
)

print(
    "Expected answer:",
    cell87_machine_validation_record[
        "expected_answer_status"
    ]
)

print(
    "Citation validation:",
    cell87_machine_validation_record[
        "citation_target_validation_status"
    ]
)

print(
    "Full 572 generation:",
    cell87_machine_validation_record[
        "full_572_generation"
    ]
)


print(
    "\nSTATUS:",
    cell87_status
)


print(
    "\n=== CODE CELL 87 GATES ==="
)


for gate_name, gate_value in (
    cell87_gates.items()
):

    print(
        gate_name,
        ":",
        (
            "PASS"
            if gate_value
            else "FAIL"
        )
    )


print(
    "\nFAILED GATES:",
    [
        gate_name

        for gate_name, gate_value
        in cell87_gates.items()

        if not gate_value
    ]
)


print(
    "\nCODE CELL 87 GATE:",
    (
        "PASS"
        if cell_87_pass
        else "FAIL"
    )
)


assert cell_87_pass, (
    "Source-Agnostic revised Question failed Machine Validation. "
    "Do not perform Human Acceptance or generate an Expected Answer. "
    "Repair this SAME Code Cell 87 using the printed diagnostics."
)
```

<h3>ساخت Claim → Evidence Part → Locator Mapping برای اولین QA پذیرفته‌شده</h3>

<p>
Cell 91 تأیید کرد که Question و Expected Answer هر دو Human-Accepted هستند و مرحله‌ی
<code>Citation Target Validation</code>
اکنون مجاز است.
</p>

<p>
در این مرحله هر Claim موجود در Expected Answer به Evidence Part واقعی‌ای که آن Claim را پشتیبانی می‌کند متصل می‌شود و Locator نیز مستقیماً از همان Evidence Part برداشته می‌شود.
</p>

<h4>اصل مهم</h4>

<ul>
<li>هیچ Locator جدیدی ساخته یا حدس زده نمی‌شود؛</li>
<li>Locator فقط از Metadata واقعی Target Evidence خوانده می‌شود؛</li>
<li>هر Claim باید حداقل یک Evidence Part داشته باشد که تمام Evidence Anchorهای لازم را پوشش دهد؛</li>
<li>Evidence Text و Locator اصلی Mutation نمی‌شوند؛</li>
<li>این Mapping کاملاً Internal است و Source Opacity در User-Facing Answer همچنان حفظ می‌شود.</li>
</ul>

<p>
این Cell فقط Mapping و Structural Validation آن را انجام می‌دهد. Final Citation Acceptance هنوز در مرحله‌ی بعد انجام خواهد شد.

اکنون هدف این مرحله این است که همین فرآیند به یک Workflow Contract تکرارپذیر تبدیل شود تا برای Pilotهای باقی‌مانده مجبور نباشیم منطق را دوباره از صفر طراحی کنیم.
</p>

<h4>اصل مهم</h4>

<p>
Human Review همچنان طبق Contract فعلی قبل از Acceptance الزامی است، اما تمام بخش‌های Machine Generation، Validation، Grounding، Source Opacity، Provenance و Citation Mapping باید با یک منطق مشترک و تکرارپذیر اجرا شوند.
</p>

<p>
بنابراین از اینجا به بعد قرار نیست برای هر Pilot یک Pipeline جدید طراحی شود؛ فقط همان State Machine روی Case بعدی اجرا خواهد شد.
</p>

<h4>این Cell چه کاری انجام می‌دهد؟</h4>

<ul>
<li>اولین Pilot تکمیل‌شده را به‌عنوان Reference Workflow ثبت می‌کند؛</li>
<li>مراحل استاندارد QA Lifecycle را Freeze می‌کند؛</li>
<li>Source-Opacity Rule را به‌عنوان Rule سراسری حفظ می‌کند؛</li>
<li>Human Gateهای الزامی را مشخص می‌کند؛</li>
<li>Queue دقیق Pilotهای باقی‌مانده را از Runtime موجود استخراج می‌کند؛</li>
<li>هیچ Question، Answer یا Citation جدیدی تولید نمی‌کند؛</li>
<li>هیچ Pilot موجودی Mutation نمی‌شود؛</li>
<li>Full 572-case Generation همچنان Blocked باقی می‌ماند.</li>
</ul>

<p>
اگر این Cell PASS شود، مرحله‌ی بعد می‌تواند Case بعدی را مستقیماً از Queue وارد همین Workflow استاندارد کند.
</p>

</div>

```python
# Code Cell 95

# ============================================================
# Freeze reusable Pilot QA Workflow Contract
# and derive remaining Pilot Queue.
#
# No new question generation.
# No expected-answer generation.
# No citation generation.
# No pilot mutation.
# No artifact write.
# ============================================================


# ============================================================
# 1) Prerequisites
# ============================================================

cell95_prerequisite_failures = []


if (
    globals().get(
        "cell_94_pass"
    )
    is not True
):

    cell95_prerequisite_failures.append(
        "CELL94_NOT_PASS"
    )


if (
    globals().get(
        "cell94_status"
    )
    !=
    "FIRST_PILOT_QA_COMPLETE_QUESTION_ANSWER_CITATION_ACCEPTED"
):

    cell95_prerequisite_failures.append(
        {
            "reason":
                "CELL94_STATUS_UNEXPECTED",

            "actual":
                globals().get(
                    "cell94_status"
                ),
        }
    )


CELL95_REQUIRED_RUNTIME_OBJECTS = [
    "cell94_completed_pilot_qa_record",
    "cell94_completed_pilot_qa_record_sha256",
    "question_drafting_pilot_packets",
    "question_drafting_pilot_by_id",
    "PILOT_QUESTION_AUTHORING_MODE_V1",
    "QUESTION_DRAFT_ACCEPTANCE_RUBRIC_V1",
    "QUESTION_USER_FACING_SOURCE_OPACITY_RULE_V1",
]


cell95_missing_runtime_objects = [
    name

    for name
    in CELL95_REQUIRED_RUNTIME_OBJECTS

    if (
        name
        not in
        globals()
    )
]


if (
    cell95_missing_runtime_objects
):

    cell95_prerequisite_failures.append(
        {
            "reason":
                "REQUIRED_RUNTIME_OBJECTS_MISSING",

            "objects":
                cell95_missing_runtime_objects,
        }
    )


# ============================================================
# 2) Helper
# ============================================================

def cell95_sha(value):

    return hashlib.sha256(
        stable_json_key(
            value
        ).encode(
            "utf-8"
        )
    ).hexdigest()


# ============================================================
# 3) Freeze state BEFORE contract / queue derivation
# ============================================================

CELL95_ACTIVE_STATE_OBJECT_NAMES = [
    "golden_case_records",
    "case_groundability_precheck",
    "case_task_design",
    "task_draft_plan_rows",
    "target_evidence_records",
    "pilot_selection_rows",
    "golden_case_records_by_id",
    "task_draft_plans",
    "target_evidence_by_case_id",
    "target_evidence_by_task_draft_id",
]


cell95_active_snapshot_before = (
    cell95_sha(
        {
            name:
                globals()[name]

            for name
            in CELL95_ACTIVE_STATE_OBJECT_NAMES

            if (
                name
                in
                globals()
            )
        }
    )
)


cell95_pilot_snapshot_before = (
    cell95_sha(
        {
            "question_drafting_pilot_packets":
                question_drafting_pilot_packets,

            "question_drafting_pilot_by_id":
                question_drafting_pilot_by_id,
        }
    )
)


# ============================================================
# 4) Resolve completed reference Pilot
# ============================================================

cell95_reference_completed_pilot_id = (
    cell94_completed_pilot_qa_record[
        "pilot_case_id"
    ]
)


cell95_reference_failures = []


if (
    cell94_completed_pilot_qa_record.get(
        "pilot_qa_completion_status"
    )
    !=
    "COMPLETE"
):

    cell95_reference_failures.append(
        "REFERENCE_PILOT_NOT_COMPLETE"
    )


if (
    cell94_completed_pilot_qa_record.get(
        "question_acceptance_status"
    )
    !=
    "ACCEPTED"
):

    cell95_reference_failures.append(
        "REFERENCE_QUESTION_NOT_ACCEPTED"
    )


if (
    cell94_completed_pilot_qa_record.get(
        "expected_answer_acceptance_status"
    )
    !=
    "ACCEPTED"
):

    cell95_reference_failures.append(
        "REFERENCE_EXPECTED_ANSWER_NOT_ACCEPTED"
    )


if (
    cell94_completed_pilot_qa_record.get(
        "citation_acceptance_status"
    )
    !=
    "ACCEPTED"
):

    cell95_reference_failures.append(
        "REFERENCE_CITATION_NOT_ACCEPTED"
    )


if (
    cell95_reference_completed_pilot_id
    not in
    question_drafting_pilot_by_id
):

    cell95_reference_failures.append(
        "REFERENCE_PILOT_NOT_FOUND_IN_FROZEN_PILOT_RUNTIME"
    )


# ============================================================
# 5) Freeze reusable lifecycle
# ============================================================

PILOT_QA_WORKFLOW_CONTRACT_V1 = {
    "version":
        "pilot_qa_workflow_contract_v1",

    "scope":
        "QUESTION_DRAFTING_PILOT_CASES",

    "reference_completed_pilot_id":
        cell95_reference_completed_pilot_id,

    "reference_completed_pilot_qa_sha256":
        cell94_completed_pilot_qa_record_sha256,

    "product_role":
        "FINANCIAL_LITERACY_ADVISOR",

    "user_facing_source_policy":
        "SOURCE_AGNOSTIC",

    "internal_provenance_policy":
        "RETAIN_AND_VALIDATE",

    "external_knowledge_policy":
        "FORBIDDEN",

    "lifecycle_stages": [
        "STAGE_01_SELECT_PILOT_PACKET",
        "STAGE_02_CREATE_QUESTION_DRAFT",
        "STAGE_03_MACHINE_VALIDATE_QUESTION",
        "STAGE_04_HUMAN_ADJUDICATE_QUESTION",
        "STAGE_05_GENERATE_EXPECTED_ANSWER",
        "STAGE_06_MACHINE_VALIDATE_EXPECTED_ANSWER",
        "STAGE_07_HUMAN_ADJUDICATE_EXPECTED_ANSWER",
        "STAGE_08_MAP_CLAIMS_TO_EVIDENCE_AND_LOCATOR",
        "STAGE_09_MACHINE_VALIDATE_CITATION_MAPPING",
        "STAGE_10_HUMAN_ADJUDICATE_CITATION_MAPPING",
        "STAGE_11_CLOSE_COMPLETED_PILOT_QA",
    ],

    "human_gates": {
        "question_acceptance":
            "REQUIRED",

        "expected_answer_acceptance":
            "REQUIRED",

        "citation_mapping_acceptance":
            "REQUIRED",
    },

    "machine_validation_requirements": {
        "question":
            "FROZEN_ACCEPTANCE_RUBRIC_PLUS_SOURCE_OPACITY",

        "expected_answer":
            "FULL_GROUNDING_COMPLETENESS_NO_EXTERNAL_KNOWLEDGE",

        "citation_mapping":
            "CLAIM_EVIDENCE_LOCATOR_IDENTITY_AND_AUTHORITY_VALIDATION",
    },

    "question_revision_policy":
        "CREATE_NEW_REVISION_ID_PRESERVE_LINEAGE",

    "source_opacity_rule_version":
        QUESTION_USER_FACING_SOURCE_OPACITY_RULE_V1[
            "version"
        ],

    "question_acceptance_rubric_version":
        QUESTION_DRAFT_ACCEPTANCE_RUBRIC_V1[
            "version"
        ],

    "automatic_acceptance":
        False,

    "full_population_generation":
        "BLOCKED_PENDING_FULL_PILOT_WORKFLOW_VALIDATION",
}


cell95_workflow_contract_sha256 = (
    cell95_sha(
        PILOT_QA_WORKFLOW_CONTRACT_V1
    )
)


# ============================================================
# 6) Derive exact Pilot universe
# ============================================================

cell95_pilot_scope_expected = (
    PILOT_QUESTION_AUTHORING_MODE_V1.get(
        "pilot_scope"
    )
)


cell95_pilot_ids_in_runtime_order = [
    packet[
        "pilot_case_id"
    ]

    for packet
    in question_drafting_pilot_packets
]


cell95_pilot_universe_failures = []


if not (
    isinstance(
        cell95_pilot_scope_expected,
        int,
    )

    and

    cell95_pilot_scope_expected
    >
    0
):

    cell95_pilot_universe_failures.append(
        {
            "reason":
                "FROZEN_PILOT_SCOPE_INVALID",

            "actual":
                cell95_pilot_scope_expected,
        }
    )


if (
    len(
        cell95_pilot_ids_in_runtime_order
    )
    !=
    cell95_pilot_scope_expected
):

    cell95_pilot_universe_failures.append(
        {
            "reason":
                "PILOT_RUNTIME_COUNT_DIFFERS_FROM_FROZEN_SCOPE",

            "runtime_count":
                len(
                    cell95_pilot_ids_in_runtime_order
                ),

            "frozen_scope":
                cell95_pilot_scope_expected,
        }
    )


if (
    len(
        set(
            cell95_pilot_ids_in_runtime_order
        )
    )
    !=
    len(
        cell95_pilot_ids_in_runtime_order
    )
):

    cell95_pilot_universe_failures.append(
        "PILOT_RUNTIME_HAS_DUPLICATE_PILOT_CASE_IDS"
    )


if (
    set(
        cell95_pilot_ids_in_runtime_order
    )
    !=
    set(
        question_drafting_pilot_by_id.keys()
    )
):

    cell95_pilot_universe_failures.append(
        "PILOT_LIST_AND_PILOT_BY_ID_KEYSET_DIFFER"
    )


# ============================================================
# 7) Derive remaining queue
# ============================================================

cell95_remaining_pilot_ids = [
    pilot_case_id

    for pilot_case_id
    in cell95_pilot_ids_in_runtime_order

    if (
        pilot_case_id
        !=
        cell95_reference_completed_pilot_id
    )
]


cell95_completed_reference_occurrences = (
    cell95_pilot_ids_in_runtime_order.count(
        cell95_reference_completed_pilot_id
    )
)


cell95_expected_remaining_count = (
    (
        cell95_pilot_scope_expected
        -
        1
    )

    if (
        isinstance(
            cell95_pilot_scope_expected,
            int,
        )
    )

    else None
)


cell95_queue_failures = []


if (
    cell95_completed_reference_occurrences
    !=
    1
):

    cell95_queue_failures.append(
        {
            "reason":
                "COMPLETED_REFERENCE_PILOT_OCCURRENCE_NOT_EXACTLY_ONE",

            "actual":
                cell95_completed_reference_occurrences,
        }
    )


if (
    cell95_expected_remaining_count
    is not None

    and

    len(
        cell95_remaining_pilot_ids
    )
    !=
    cell95_expected_remaining_count
):

    cell95_queue_failures.append(
        {
            "reason":
                "REMAINING_QUEUE_COUNT_UNEXPECTED",

            "expected":
                cell95_expected_remaining_count,

            "actual":
                len(
                    cell95_remaining_pilot_ids
                ),
        }
    )


if (
    cell95_reference_completed_pilot_id
    in
    cell95_remaining_pilot_ids
):

    cell95_queue_failures.append(
        "COMPLETED_REFERENCE_PILOT_STILL_PRESENT_IN_REMAINING_QUEUE"
    )


# ============================================================
# 8) Build remaining Pilot queue records
# ============================================================

cell95_remaining_pilot_queue = []


for queue_index, pilot_case_id in enumerate(
    cell95_remaining_pilot_ids,
    start=1,
):

    packet = (
        question_drafting_pilot_by_id[
            pilot_case_id
        ]
    )


    cell95_remaining_pilot_queue.append(
        {
            "queue_index":
                queue_index,

            "pilot_case_id":
                pilot_case_id,

            "golden_case_id":
                packet.get(
                    "golden_case_id"
                ),

            "source_type":
                packet.get(
                    "source_type"
                ),

            "stratum":
                packet.get(
                    "stratum"
                ),

            "task_family":
                packet.get(
                    "task_family"
                ),

            "question_archetype":
                packet.get(
                    "question_archetype"
                ),

            "target_evidence_id":
                packet.get(
                    "target_evidence_id"
                ),

            "workflow_status":
                "PENDING_STANDARD_PILOT_QA_WORKFLOW",

            "question_status":
                "NOT_STARTED",

            "expected_answer_status":
                "BLOCKED_PENDING_QUESTION_ACCEPTANCE",

            "citation_status":
                "BLOCKED_PENDING_EXPECTED_ANSWER_ACCEPTANCE",
        }
    )


cell95_remaining_pilot_queue_sha256 = (
    cell95_sha(
        cell95_remaining_pilot_queue
    )
)


# ============================================================
# 9) Select next Pilot deterministically
#
# Selection only.
# No generation.
# ============================================================

cell95_next_pilot_queue_entry = (
    cell95_remaining_pilot_queue[0]

    if (
        len(
            cell95_remaining_pilot_queue
        )
        >
        0
    )

    else None
)


cell95_next_pilot_id = (
    cell95_next_pilot_queue_entry[
        "pilot_case_id"
    ]

    if (
        cell95_next_pilot_queue_entry
        is not None
    )

    else None
)


# ============================================================
# 10) Workflow state summary
# ============================================================

cell95_pilot_workflow_state = {
    "state_version":
        "pilot_qa_workflow_state_v1",

    "workflow_contract_version":
        PILOT_QA_WORKFLOW_CONTRACT_V1[
            "version"
        ],

    "workflow_contract_sha256":
        cell95_workflow_contract_sha256,

    "pilot_scope":
        cell95_pilot_scope_expected,

    "completed_pilot_count":
        1,

    "completed_reference_pilot_id":
        cell95_reference_completed_pilot_id,

    "remaining_pilot_count":
        len(
            cell95_remaining_pilot_queue
        ),

    "remaining_queue_sha256":
        cell95_remaining_pilot_queue_sha256,

    "next_pilot_id":
        cell95_next_pilot_id,

    "next_action":
        "STAGE_NEXT_PILOT_PACKET_FOR_STANDARD_WORKFLOW",

    "full_572_generation":
        "BLOCKED_PENDING_FULL_PILOT_WORKFLOW_VALIDATION",
}


cell95_pilot_workflow_state_sha256 = (
    cell95_sha(
        cell95_pilot_workflow_state
    )
)


# ============================================================
# 11) Freeze state AFTER contract / queue derivation
# ============================================================

cell95_active_snapshot_after = (
    cell95_sha(
        {
            name:
                globals()[name]

            for name
            in CELL95_ACTIVE_STATE_OBJECT_NAMES

            if (
                name
                in
                globals()
            )
        }
    )
)


cell95_pilot_snapshot_after = (
    cell95_sha(
        {
            "question_drafting_pilot_packets":
                question_drafting_pilot_packets,

            "question_drafting_pilot_by_id":
                question_drafting_pilot_by_id,
        }
    )
)


# ============================================================
# 12) Final gates
# ============================================================

cell95_gates = {
    "cell94_is_pass":
        (
            globals().get(
                "cell_94_pass"
            )
            is True
        ),

    "reference_completed_pilot_is_valid":
        (
            len(
                cell95_reference_failures
            )
            ==
            0
        ),

    "reusable_workflow_contract_is_frozen":
        (
            PILOT_QA_WORKFLOW_CONTRACT_V1[
                "automatic_acceptance"
            ]
            is False

            and

            PILOT_QA_WORKFLOW_CONTRACT_V1[
                "user_facing_source_policy"
            ]
            ==
            "SOURCE_AGNOSTIC"

            and

            PILOT_QA_WORKFLOW_CONTRACT_V1[
                "internal_provenance_policy"
            ]
            ==
            "RETAIN_AND_VALIDATE"
        ),

    "three_human_acceptance_gates_are_preserved":
        (
            PILOT_QA_WORKFLOW_CONTRACT_V1[
                "human_gates"
            ]
            ==
            {
                "question_acceptance":
                    "REQUIRED",

                "expected_answer_acceptance":
                    "REQUIRED",

                "citation_mapping_acceptance":
                    "REQUIRED",
            }
        ),

    "pilot_universe_matches_frozen_scope":
        (
            len(
                cell95_pilot_universe_failures
            )
            ==
            0
        ),

    "completed_reference_removed_exactly_once":
        (
            cell95_completed_reference_occurrences
            ==
            1

            and

            cell95_reference_completed_pilot_id
            not in
            cell95_remaining_pilot_ids
        ),

    "remaining_queue_count_is_exact":
        (
            cell95_expected_remaining_count
            is not None

            and

            len(
                cell95_remaining_pilot_queue
            )
            ==
            cell95_expected_remaining_count
        ),

    "every_remaining_queue_entry_is_pending":
        all(
            row[
                "workflow_status"
            ]
            ==
            "PENDING_STANDARD_PILOT_QA_WORKFLOW"

            for row
            in cell95_remaining_pilot_queue
        ),

    "next_pilot_is_selected_without_generation":
        (
            cell95_next_pilot_queue_entry
            is not None

            and

            cell95_next_pilot_id
            ==
            cell95_remaining_pilot_queue[0][
                "pilot_case_id"
            ]

            and

            cell95_next_pilot_queue_entry[
                "question_status"
            ]
            ==
            "NOT_STARTED"
        ),

    "full_572_generation_remains_blocked":
        (
            cell95_pilot_workflow_state[
                "full_572_generation"
            ]
            ==
            "BLOCKED_PENDING_FULL_PILOT_WORKFLOW_VALIDATION"
        ),

    "active_state_remains_byte_stable":
        (
            cell95_active_snapshot_before
            ==
            cell95_active_snapshot_after
        ),

    "pilot_runtime_remains_byte_stable":
        (
            cell95_pilot_snapshot_before
            ==
            cell95_pilot_snapshot_after
        ),

    "no_new_question_generated":
        (
            cell95_next_pilot_queue_entry[
                "question_status"
            ]
            ==
            "NOT_STARTED"
        ),

    "no_artifact_write":
        True,
}


cell_95_pass = all(
    cell95_gates.values()
)


cell95_status = (
    "REUSABLE_PILOT_QA_WORKFLOW_FROZEN_REMAINING_QUEUE_READY"

    if (
        cell_95_pass
    )

    else
    "REUSABLE_PILOT_QA_WORKFLOW_OR_QUEUE_VALIDATION_FAILED"
)


# ============================================================
# 13) Diagnostics
# ============================================================

print(
    "=== REUSABLE PILOT QA WORKFLOW + REMAINING QUEUE ==="
)


print(
    "\n=== REFERENCE COMPLETED PILOT ==="
)

print(
    "Pilot ID:",
    cell95_reference_completed_pilot_id
)

print(
    "Completion status:",
    cell94_completed_pilot_qa_record[
        "pilot_qa_completion_status"
    ]
)

print(
    "Reference failures:",
    cell95_reference_failures
)


print(
    "\n=== WORKFLOW CONTRACT ==="
)

print(
    "Version:",
    PILOT_QA_WORKFLOW_CONTRACT_V1[
        "version"
    ]
)

print(
    "Lifecycle stage count:",
    len(
        PILOT_QA_WORKFLOW_CONTRACT_V1[
            "lifecycle_stages"
        ]
    )
)

print(
    "Human gates:",
    PILOT_QA_WORKFLOW_CONTRACT_V1[
        "human_gates"
    ]
)

print(
    "Source policy:",
    PILOT_QA_WORKFLOW_CONTRACT_V1[
        "user_facing_source_policy"
    ]
)

print(
    "Internal provenance:",
    PILOT_QA_WORKFLOW_CONTRACT_V1[
        "internal_provenance_policy"
    ]
)

print(
    "Workflow contract SHA256:",
    cell95_workflow_contract_sha256
)


print(
    "\n=== PILOT UNIVERSE ==="
)

print(
    "Frozen pilot scope:",
    cell95_pilot_scope_expected
)

print(
    "Runtime pilot count:",
    len(
        cell95_pilot_ids_in_runtime_order
    )
)

print(
    "Universe failures:",
    cell95_pilot_universe_failures
)


print(
    "\n=== REMAINING QUEUE ==="
)

print(
    "Completed count:",
    cell95_pilot_workflow_state[
        "completed_pilot_count"
    ]
)

print(
    "Remaining count:",
    cell95_pilot_workflow_state[
        "remaining_pilot_count"
    ]
)

print(
    "Queue failures:",
    cell95_queue_failures
)

print(
    "Queue SHA256:",
    cell95_remaining_pilot_queue_sha256
)


print(
    "\n=== NEXT PILOT ==="
)

print(
    "Pilot ID:",
    cell95_next_pilot_id
)

print(
    "Source type:",
    cell95_next_pilot_queue_entry[
        "source_type"
    ]
)

print(
    "Stratum:",
    cell95_next_pilot_queue_entry[
        "stratum"
    ]
)

print(
    "Task family:",
    cell95_next_pilot_queue_entry[
        "task_family"
    ]
)

print(
    "Question archetype:",
    cell95_next_pilot_queue_entry[
        "question_archetype"
    ]
)

print(
    "Question status:",
    cell95_next_pilot_queue_entry[
        "question_status"
    ]
)


print(
    "\n=== DOWNSTREAM ==="
)

print(
    "Next action:",
    cell95_pilot_workflow_state[
        "next_action"
    ]
)

print(
    "Full 572 generation:",
    cell95_pilot_workflow_state[
        "full_572_generation"
    ]
)


print(
    "\nWorkflow state SHA256:",
    cell95_pilot_workflow_state_sha256
)


print(
    "\nSTATUS:",
    cell95_status
)


print(
    "\n=== CODE CELL 95 GATES ==="
)


for gate_name, gate_value in (
    cell95_gates.items()
):

    print(
        gate_name,
        ":",
        (
            "PASS"
            if gate_value
            else "FAIL"
        )
    )


print(
    "\nFAILED GATES:",
    [
        gate_name

        for gate_name, gate_value
        in cell95_gates.items()

        if not gate_value
    ]
)


print(
    "\nCODE CELL 95 GATE:",
    (
        "PASS"
        if cell_95_pass
        else "FAIL"
    )
)


assert cell_95_pass, (
    "Reusable Pilot QA workflow freeze or remaining queue "
    "derivation failed. Do not start the next Pilot. "
    "Repair this SAME Code Cell 95 using the printed diagnostics."
)
```

```python
# Code Cell 97

# ============================================================
# Stage 02 — Create Question Draft for the next Pilot.
#
# No machine acceptance.
# No human acceptance.
# No expected-answer generation.
# No citation mapping.
# No runtime mutation.
# No artifact write.
# ============================================================


# ============================================================
# 1) Prerequisites
# ============================================================

cell97_prerequisite_failures = []


if (
    globals().get(
        "cell_96_pass"
    )
    is not True
):

    cell97_prerequisite_failures.append(
        "CELL96_NOT_PASS"
    )


if (
    globals().get(
        "cell96_status"
    )
    !=
    "NEXT_PILOT_STAGED_FOR_STANDARD_WORKFLOW_NO_GENERATION"
):

    cell97_prerequisite_failures.append(
        {
            "reason":
                "CELL96_STATUS_UNEXPECTED",

            "actual":
                globals().get(
                    "cell96_status"
                ),
        }
    )


CELL97_REQUIRED_RUNTIME_OBJECTS = [
    "cell96_next_pilot_standard_workflow_stage",
    "cell96_staged_packet_sha256",
    "PILOT_QA_WORKFLOW_CONTRACT_V1",
    "QUESTION_DRAFT_RECORD_SCHEMA_V1",
    "QUESTION_DRAFT_ID_CONTRACT_V1",
    "PILOT_QUESTION_AUTHORING_MODE_V1",
    "QUESTION_USER_FACING_SOURCE_OPACITY_RULE_V1",
    "build_question_draft_id",
]


cell97_missing_runtime_objects = [
    name

    for name
    in CELL97_REQUIRED_RUNTIME_OBJECTS

    if (
        name
        not in
        globals()
    )
]


if (
    cell97_missing_runtime_objects
):

    cell97_prerequisite_failures.append(
        {
            "reason":
                "REQUIRED_RUNTIME_OBJECTS_MISSING",

            "objects":
                cell97_missing_runtime_objects,
        }
    )


CELL97_EXPECTED_STAGING_SHA256 = (
    "1edc7d899bc1dafc0408be14ecbc121cd"
    "1bdffbf8157e72c43367aaa76f55d75"
)


if (
    globals().get(
        "cell96_staged_packet_sha256"
    )
    !=
    CELL97_EXPECTED_STAGING_SHA256
):

    cell97_prerequisite_failures.append(
        {
            "reason":
                "CELL96_STAGING_SHA256_MISMATCH",

            "expected":
                CELL97_EXPECTED_STAGING_SHA256,

            "actual":
                globals().get(
                    "cell96_staged_packet_sha256"
                ),
        }
    )


# ============================================================
# 2) Helper
# ============================================================

def cell97_sha(value):

    return hashlib.sha256(
        stable_json_key(
            value
        ).encode(
            "utf-8"
        )
    ).hexdigest()


# ============================================================
# 3) Resolve frozen workflow stage
# ============================================================

cell97_workflow_failures = []


cell97_lifecycle_stages = (
    PILOT_QA_WORKFLOW_CONTRACT_V1.get(
        "lifecycle_stages",
        [],
    )
)


if not (
    len(
        cell97_lifecycle_stages
    )
    >=
    3

    and

    cell97_lifecycle_stages[
        1
    ]
    ==
    "STAGE_02_CREATE_QUESTION_DRAFT"

    and

    cell97_lifecycle_stages[
        2
    ]
    ==
    "STAGE_03_MACHINE_VALIDATE_QUESTION"
):

    cell97_workflow_failures.append(
        {
            "reason":
                "FROZEN_WORKFLOW_STAGE_SEQUENCE_UNEXPECTED",

            "actual":
                cell97_lifecycle_stages,
        }
    )


if (
    PILOT_QA_WORKFLOW_CONTRACT_V1.get(
        "automatic_acceptance"
    )
    is not False
):

    cell97_workflow_failures.append(
        "AUTOMATIC_ACCEPTANCE_NOT_FALSE"
    )


if (
    PILOT_QA_WORKFLOW_CONTRACT_V1.get(
        "external_knowledge_policy"
    )
    !=
    "FORBIDDEN"
):

    cell97_workflow_failures.append(
        "EXTERNAL_KNOWLEDGE_NOT_FORBIDDEN"
    )


# ============================================================
# 4) Freeze protected runtime BEFORE drafting
# ============================================================

CELL97_ACTIVE_STATE_OBJECT_NAMES = [
    "golden_case_records",
    "case_groundability_precheck",
    "case_task_design",
    "task_draft_plan_rows",
    "target_evidence_records",
    "pilot_selection_rows",
    "golden_case_records_by_id",
    "task_draft_plans",
    "target_evidence_by_case_id",
    "target_evidence_by_task_draft_id",
]


cell97_active_snapshot_before = (
    cell97_sha(
        {
            name:
                globals()[name]

            for name
            in CELL97_ACTIVE_STATE_OBJECT_NAMES

            if (
                name
                in
                globals()
            )
        }
    )
)


cell97_pilot_snapshot_before = (
    cell97_sha(
        {
            "question_drafting_pilot_packets":
                question_drafting_pilot_packets,

            "question_drafting_pilot_by_id":
                question_drafting_pilot_by_id,
        }
    )
)


cell97_queue_snapshot_before = (
    cell97_sha(
        cell95_remaining_pilot_queue
    )
)


cell97_workflow_state_snapshot_before = (
    cell97_sha(
        cell95_pilot_workflow_state
    )
)


cell97_stage_snapshot_before = (
    cell97_sha(
        cell96_next_pilot_standard_workflow_stage
    )
)


# ============================================================
# 5) Frozen staged input
# ============================================================

cell97_packet = (
    cell96_next_pilot_standard_workflow_stage
)


cell97_packet_failures = []


CELL97_EXPECTED_PILOT = {
    "pilot_case_id":
        "question-pilot:532a8b9d45efb8b3a32c4a62f096bb7a",

    "source_type":
        "website",

    "stratum":
        "direct_author_long",

    "task_family":
        "LONG_DIRECT_PRIMARY_SOURCE_RETRIEVAL",

    "question_archetype":
        "EVIDENCE_COMPLETENESS",

    "authority_expectation":
        "PRIMARY_DIRECT_AUTHOR",

    "context_expectation":
        "SUFFICIENT_WITHOUT_EXCESS",

    "evidence_scope":
        "SINGLE_LONG_SOURCE_DOCUMENT",
}


for field_name, expected_value in (
    CELL97_EXPECTED_PILOT.items()
):

    actual_value = (
        cell97_packet.get(
            field_name
        )
    )


    if (
        actual_value
        !=
        expected_value
    ):

        cell97_packet_failures.append(
            {
                "reason":
                    "STAGED_PILOT_FIELD_MISMATCH",

                "field":
                    field_name,

                "expected":
                    expected_value,

                "actual":
                    actual_value,
            }
        )


if (
    cell97_packet.get(
        "question_generation_status"
    )
    !=
    "NOT_STARTED"
):

    cell97_packet_failures.append(
        "QUESTION_GENERATION_ALREADY_STARTED"
    )


if (
    cell97_packet.get(
        "question_text"
    )
    is not None
):

    cell97_packet_failures.append(
        "STAGED_PACKET_ALREADY_HAS_QUESTION_TEXT"
    )


# ============================================================
# 6) Evidence integrity and authoring boundary
# ============================================================

cell97_target_parts = (
    cell97_packet.get(
        "target_text_parts",
        [],
    )
)


cell97_evidence_failures = []


if (
    len(
        cell97_target_parts
    )
    !=
    1
):

    cell97_evidence_failures.append(
        {
            "reason":
                "TARGET_PART_COUNT_UNEXPECTED",

            "expected":
                1,

            "actual":
                len(
                    cell97_target_parts
                ),
        }
    )


cell97_joined_evidence = "\n".join(
    part.get(
        "text",
        "",
    )

    for part
    in cell97_target_parts
)


for part_index, part in enumerate(
    cell97_target_parts
):

    text = (
        part.get(
            "text"
        )
    )


    if not (
        isinstance(
            text,
            str,
        )

        and

        text.strip()
    ):

        cell97_evidence_failures.append(
            {
                "reason":
                    "TARGET_TEXT_EMPTY",

                "part_index":
                    part_index,
            }
        )

        continue


    recomputed_sha = hashlib.sha256(
        text.encode(
            "utf-8"
        )
    ).hexdigest()


    if (
        recomputed_sha
        !=
        part.get(
            "text_sha256"
        )
    ):

        cell97_evidence_failures.append(
            {
                "reason":
                    "TARGET_TEXT_SHA256_MISMATCH",

                "part_index":
                    part_index,
            }
        )


    attribution = (
        part.get(
            "attribution"
        )
        or
        {}
    )


    if not (
        attribution.get(
            "content_authority_type"
        )
        ==
        "direct_author"

        and

        attribution.get(
            "attribution_status"
        )
        ==
        "policy_validated"
    ):

        cell97_evidence_failures.append(
            {
                "reason":
                    "DIRECT_AUTHOR_AUTHORITY_NOT_VALIDATED",

                "part_index":
                    part_index,
            }
        )


CELL97_EXPECTED_EVIDENCE_ANCHORS = [
    "خرج",
    "پس‌انداز",
    "شوک‌های درآمدی",
    "هزین‌های",
    "اعتبار",
    "بدهی",
    "سرمایه‌گذاری",
    "مدیریت ریسک",
    "تصمیم‌گیری مالی",
    "چهارده مفهوم سواد مالی",
]


cell97_missing_evidence_anchors = [
    anchor

    for anchor
    in CELL97_EXPECTED_EVIDENCE_ANCHORS

    if (
        anchor
        not in
        cell97_joined_evidence
    )
]


if (
    cell97_missing_evidence_anchors
):

    cell97_evidence_failures.append(
        {
            "reason":
                "EXPECTED_QUESTION_SCOPE_NOT_FULLY_SUPPORTED_BY_TARGET",

            "missing":
                cell97_missing_evidence_anchors,
        }
    )


# ============================================================
# 7) Create source-agnostic Question Draft
# ============================================================

CELL97_QUESTION_TEXT = (
    "چه موضوعات سواد مالی به‌طور صریح در کنار هم مطرح شده‌اند "
    "و تعداد کل مفاهیم سواد مالی چند مورد اعلام شده است؟"
)


cell97_question_text_sha256 = (
    hashlib.sha256(
        CELL97_QUESTION_TEXT.encode(
            "utf-8"
        )
    ).hexdigest()
)


# ============================================================
# 8) Draft-authoring preflight
#
# Full machine validation belongs to Cell 98.
# ============================================================

cell97_authoring_preflight = {}


cell97_authoring_preflight[
    "natural_nonempty_question"
] = (
    isinstance(
        CELL97_QUESTION_TEXT,
        str,
    )

    and

    len(
        CELL97_QUESTION_TEXT.strip()
    )
    >=
    20

    and

    CELL97_QUESTION_TEXT.endswith(
        "؟"
    )
)


CELL97_INTERNAL_MARKERS = [
    "golden-case:",
    "task-draft:",
    "target-evidence:",
    "question-pilot:",
    "question-draft:",
    "website:",
    "document_id",
    "source_id",
    "sha256",
]


cell97_authoring_preflight[
    "no_internal_pipeline_metadata"
] = not any(
    marker
    in
    CELL97_QUESTION_TEXT

    for marker
    in CELL97_INTERNAL_MARKERS
)


cell97_authoring_preflight[
    "no_url_or_locator"
] = not any(
    marker
    in
    CELL97_QUESTION_TEXT

    for marker
    in [
        "http://",
        "https://",
        "www.",
    ]
)


cell97_attribution_identity_names = set()


for part in (
    cell97_target_parts
):

    attribution = (
        part.get(
            "attribution"
        )
        or
        {}
    )


    for field_name in [
        "author",
        "credited_author_name",
        "knowledge_owner",
        "editorial_supervisor_name",
    ]:

        value = (
            attribution.get(
                field_name
            )
        )


        if (
            isinstance(
                value,
                str,
            )

            and

            value.strip()
        ):

            cell97_attribution_identity_names.add(
                value.strip()
            )


cell97_source_identity_hits = [
    value

    for value
    in sorted(
        cell97_attribution_identity_names
    )

    if (
        value
        in
        CELL97_QUESTION_TEXT
    )
]


CELL97_SOURCE_SPECIFIC_TERMS = [
    "شب طلایی",
]


cell97_source_specific_hits = [
    value

    for value
    in CELL97_SOURCE_SPECIFIC_TERMS

    if (
        value
        in
        CELL97_QUESTION_TEXT
    )
]


cell97_authoring_preflight[
    "user_facing_source_identity_not_exposed"
] = (
    len(
        cell97_source_identity_hits
    )
    ==
    0

    and

    len(
        cell97_source_specific_hits
    )
    ==
    0
)


CELL97_SOURCE_REFERENCE_CUES = [
    "بر اساس",
    "به گفته",
    "طبق گفته",
    "طبق نظر",
    "از دیدگاه",
    "از نظر نویسنده",
    "از نظر دکتر",
    "در این مقاله",
    "در این منبع",
    "نویسنده",
    "منبع",
]


cell97_source_reference_cue_hits = [
    cue

    for cue
    in CELL97_SOURCE_REFERENCE_CUES

    if (
        cue
        in
        CELL97_QUESTION_TEXT
    )
]


cell97_authoring_preflight[
    "source_reference_cues_not_exposed"
] = (
    len(
        cell97_source_reference_cue_hits
    )
    ==
    0
)


CELL97_ANSWER_LEAKAGE_TERMS = [
    "خرج",
    "پس‌انداز",
    "شوک‌های درآمدی",
    "هزین‌های",
    "اعتبار",
    "بدهی",
    "سرمایه‌گذاری",
    "مدیریت ریسک",
    "تصمیم‌گیری مالی",
    "چهارده",
    "۱۴",
]


cell97_answer_leakage_hits = [
    value

    for value
    in CELL97_ANSWER_LEAKAGE_TERMS

    if (
        value
        in
        CELL97_QUESTION_TEXT
    )
]


cell97_authoring_preflight[
    "answer_not_leaked_into_question"
] = (
    len(
        cell97_answer_leakage_hits
    )
    ==
    0
)


cell97_authoring_preflight[
    "question_does_not_copy_target_text"
] = (
    CELL97_QUESTION_TEXT
    not in
    cell97_joined_evidence

    and

    cell97_joined_evidence
    not in
    CELL97_QUESTION_TEXT
)


cell97_authoring_preflight[
    "source_opacity_rule_is_frozen"
] = (
    QUESTION_USER_FACING_SOURCE_OPACITY_RULE_V1.get(
        "user_facing_question_policy"
    )
    ==
    "SOURCE_AGNOSTIC"

    and

    QUESTION_USER_FACING_SOURCE_OPACITY_RULE_V1.get(
        "internal_provenance_retention"
    )
    ==
    "REQUIRED"

    and

    QUESTION_USER_FACING_SOURCE_OPACITY_RULE_V1.get(
        "external_knowledge"
    )
    ==
    "FORBIDDEN"
)


cell97_preflight_failures = [
    gate_name

    for gate_name, gate_value
    in cell97_authoring_preflight.items()

    if not gate_value
]


# ============================================================
# 9) Resolve frozen schema and initial statuses
# ============================================================

cell97_schema_version = (
    QUESTION_DRAFT_RECORD_SCHEMA_V1.get(
        "schema_version"
    )
)


cell97_required_fields = (
    QUESTION_DRAFT_RECORD_SCHEMA_V1.get(
        "required_fields",
        [],
    )
)


cell97_initial_statuses = (
    QUESTION_DRAFT_RECORD_SCHEMA_V1.get(
        "initial_statuses",
        {},
    )
)


cell97_schema_failures = []


if (
    cell97_schema_version
    !=
    "question_draft_record_v1"
):

    cell97_schema_failures.append(
        {
            "reason":
                "QUESTION_DRAFT_SCHEMA_VERSION_UNEXPECTED",

            "actual":
                cell97_schema_version,
        }
    )


CELL97_EXPECTED_INITIAL_STATUSES = {
    "draft_status":
        "DRAFT_CREATED_PENDING_VALIDATION",

    "machine_validation_status":
        "NOT_STARTED",

    "human_adjudication_status":
        "NOT_STARTED",

    "expected_answer_status":
        "BLOCKED_PENDING_QUESTION_ACCEPTANCE",

    "citation_target_validation_status":
        "BLOCKED_PENDING_QUESTION_ACCEPTANCE",
}


if (
    cell97_initial_statuses
    !=
    CELL97_EXPECTED_INITIAL_STATUSES
):

    cell97_schema_failures.append(
        {
            "reason":
                "QUESTION_INITIAL_STATUS_CONTRACT_UNEXPECTED",

            "actual":
                cell97_initial_statuses,
        }
    )


# ============================================================
# 10) Deterministic Question Draft ID
# ============================================================

cell97_id_failures = []


cell97_question_draft_id = (
    build_question_draft_id(
        pilot_case_id=
            cell97_packet[
                "pilot_case_id"
            ],

        question_text_sha256=
            cell97_question_text_sha256,
    )
)


cell97_question_draft_id_repeat = (
    build_question_draft_id(
        pilot_case_id=
            cell97_packet[
                "pilot_case_id"
            ],

        question_text_sha256=
            cell97_question_text_sha256,
    )
)


if not (
    isinstance(
        cell97_question_draft_id,
        str,
    )

    and

    cell97_question_draft_id.startswith(
        QUESTION_DRAFT_ID_CONTRACT_V1[
            "prefix"
        ]
    )
):

    cell97_id_failures.append(
        {
            "reason":
                "QUESTION_DRAFT_ID_INVALID",

            "actual":
                cell97_question_draft_id,
        }
    )


if (
    cell97_question_draft_id
    !=
    cell97_question_draft_id_repeat
):

    cell97_id_failures.append(
        "QUESTION_DRAFT_ID_NOT_DETERMINISTIC"
    )


# ============================================================
# 11) Materialize exact frozen Question Draft record
# ============================================================

cell97_question_field_values = {
    "question_draft_id":
        cell97_question_draft_id,

    "question_draft_schema_version":
        cell97_schema_version,

    "pilot_case_id":
        cell97_packet[
            "pilot_case_id"
        ],

    "golden_case_id":
        cell97_packet[
            "golden_case_id"
        ],

    "task_draft_id":
        cell97_packet[
            "task_draft_id"
        ],

    "target_evidence_id":
        cell97_packet[
            "target_evidence_id"
        ],

    "source_type":
        cell97_packet[
            "source_type"
        ],

    "stratum":
        cell97_packet[
            "stratum"
        ],

    "task_family":
        cell97_packet[
            "task_family"
        ],

    "question_archetype":
        cell97_packet[
            "question_archetype"
        ],

    "required_dimensions":
        list(
            cell97_packet[
                "required_dimensions"
            ]
        ),

    "authority_expectation":
        cell97_packet[
            "authority_expectation"
        ],

    "context_expectation":
        cell97_packet[
            "context_expectation"
        ],

    "evidence_scope":
        cell97_packet[
            "evidence_scope"
        ],

    "question_text":
        CELL97_QUESTION_TEXT,

    "question_text_sha256":
        cell97_question_text_sha256,

    "authoring_mode":
        PILOT_QUESTION_AUTHORING_MODE_V1[
            "mode"
        ],

    "draft_status":
        cell97_initial_statuses[
            "draft_status"
        ],

    "machine_validation_status":
        cell97_initial_statuses[
            "machine_validation_status"
        ],

    "human_adjudication_status":
        cell97_initial_statuses[
            "human_adjudication_status"
        ],

    "expected_answer_status":
        cell97_initial_statuses[
            "expected_answer_status"
        ],

    "citation_target_validation_status":
        cell97_initial_statuses[
            "citation_target_validation_status"
        ],
}


cell97_unmapped_required_fields = [
    field_name

    for field_name
    in cell97_required_fields

    if (
        field_name
        not in
        cell97_question_field_values
    )
]


if (
    cell97_unmapped_required_fields
):

    cell97_schema_failures.append(
        {
            "reason":
                "UNMAPPED_REQUIRED_SCHEMA_FIELDS",

            "fields":
                cell97_unmapped_required_fields,
        }
    )


cell97_question_draft_record = (
    {
        field_name:
            cell97_question_field_values[
                field_name
            ]

        for field_name
        in cell97_required_fields
    }

    if (
        len(
            cell97_prerequisite_failures
        )
        ==
        0

        and

        len(
            cell97_workflow_failures
        )
        ==
        0

        and

        len(
            cell97_packet_failures
        )
        ==
        0

        and

        len(
            cell97_evidence_failures
        )
        ==
        0

        and

        len(
            cell97_preflight_failures
        )
        ==
        0

        and

        len(
            cell97_schema_failures
        )
        ==
        0

        and

        len(
            cell97_id_failures
        )
        ==
        0
    )

    else None
)


cell97_question_draft_record_sha256 = (
    cell97_sha(
        cell97_question_draft_record
    )

    if (
        cell97_question_draft_record
        is not None
    )

    else None
)


# ============================================================
# 12) Preserve draft lineage
# ============================================================

cell97_question_draft_lineage = (
    {
        "lineage_version":
            "standard_pilot_question_draft_lineage_v1",

        "workflow_stage":
            "STAGE_02_CREATE_QUESTION_DRAFT",

        "workflow_contract_sha256":
            cell95_workflow_contract_sha256,

        "staging_packet_sha256":
            cell96_staged_packet_sha256,

        "pilot_case_id":
            cell97_packet[
                "pilot_case_id"
            ],

        "target_evidence_id":
            cell97_packet[
                "target_evidence_id"
            ],

        "target_text_sha256":
            [
                part[
                    "text_sha256"
                ]

                for part
                in cell97_target_parts
            ],

        "source_opacity_rule_version":
            QUESTION_USER_FACING_SOURCE_OPACITY_RULE_V1[
                "version"
            ],

        "question_draft_id":
            cell97_question_draft_id,

        "question_text_sha256":
            cell97_question_text_sha256,

        "next_action":
            "STAGE_03_MACHINE_VALIDATE_QUESTION",
    }

    if (
        cell97_question_draft_record
        is not None
    )

    else None
)


cell97_question_draft_lineage_sha256 = (
    cell97_sha(
        cell97_question_draft_lineage
    )

    if (
        cell97_question_draft_lineage
        is not None
    )

    else None
)


# ============================================================
# 13) Freeze protected runtime AFTER drafting
# ============================================================

cell97_active_snapshot_after = (
    cell97_sha(
        {
            name:
                globals()[name]

            for name
            in CELL97_ACTIVE_STATE_OBJECT_NAMES

            if (
                name
                in
                globals()
            )
        }
    )
)


cell97_pilot_snapshot_after = (
    cell97_sha(
        {
            "question_drafting_pilot_packets":
                question_drafting_pilot_packets,

            "question_drafting_pilot_by_id":
                question_drafting_pilot_by_id,
        }
    )
)


cell97_queue_snapshot_after = (
    cell97_sha(
        cell95_remaining_pilot_queue
    )
)


cell97_workflow_state_snapshot_after = (
    cell97_sha(
        cell95_pilot_workflow_state
    )
)


cell97_stage_snapshot_after = (
    cell97_sha(
        cell96_next_pilot_standard_workflow_stage
    )
)


# ============================================================
# 14) Final gates
# ============================================================

cell97_gates = {
    "cell96_is_pass":
        (
            globals().get(
                "cell_96_pass"
            )
            is True
        ),

    "workflow_stage_02_is_exact":
        (
            len(
                cell97_workflow_failures
            )
            ==
            0
        ),

    "staged_pilot_is_exact":
        (
            len(
                cell97_packet_failures
            )
            ==
            0
        ),

    "target_evidence_supports_question_scope":
        (
            len(
                cell97_evidence_failures
            )
            ==
            0
        ),

    "draft_authoring_preflight_passes":
        (
            len(
                cell97_preflight_failures
            )
            ==
            0
        ),

    "question_schema_is_frozen_and_complete":
        (
            len(
                cell97_schema_failures
            )
            ==
            0
        ),

    "question_draft_id_is_valid_and_deterministic":
        (
            len(
                cell97_id_failures
            )
            ==
            0
        ),

    "question_draft_record_is_materialized":
        (
            cell97_question_draft_record
            is not None
        ),

    "question_is_source_agnostic":
        (
            cell97_authoring_preflight[
                "user_facing_source_identity_not_exposed"
            ]

            and

            cell97_authoring_preflight[
                "source_reference_cues_not_exposed"
            ]
        ),

    "question_does_not_leak_answer":
        (
            cell97_authoring_preflight[
                "answer_not_leaked_into_question"
            ]
        ),

    "machine_validation_not_started":
        (
            cell97_question_draft_record
            is not None

            and

            cell97_question_draft_record[
                "machine_validation_status"
            ]
            ==
            "NOT_STARTED"
        ),

    "human_adjudication_not_started":
        (
            cell97_question_draft_record
            is not None

            and

            cell97_question_draft_record[
                "human_adjudication_status"
            ]
            ==
            "NOT_STARTED"
        ),

    "expected_answer_remains_blocked":
        (
            cell97_question_draft_record
            is not None

            and

            cell97_question_draft_record[
                "expected_answer_status"
            ]
            ==
            "BLOCKED_PENDING_QUESTION_ACCEPTANCE"
        ),

    "citation_validation_remains_blocked":
        (
            cell97_question_draft_record
            is not None

            and

            cell97_question_draft_record[
                "citation_target_validation_status"
            ]
            ==
            "BLOCKED_PENDING_QUESTION_ACCEPTANCE"
        ),

    "lineage_is_preserved":
        (
            cell97_question_draft_lineage
            is not None

            and

            cell97_question_draft_lineage[
                "next_action"
            ]
            ==
            "STAGE_03_MACHINE_VALIDATE_QUESTION"
        ),

    "active_state_remains_byte_stable":
        (
            cell97_active_snapshot_before
            ==
            cell97_active_snapshot_after
        ),

    "pilot_runtime_remains_byte_stable":
        (
            cell97_pilot_snapshot_before
            ==
            cell97_pilot_snapshot_after
        ),

    "remaining_queue_remains_byte_stable":
        (
            cell97_queue_snapshot_before
            ==
            cell97_queue_snapshot_after
        ),

    "workflow_state_remains_byte_stable":
        (
            cell97_workflow_state_snapshot_before
            ==
            cell97_workflow_state_snapshot_after
        ),

    "staging_packet_remains_byte_stable":
        (
            cell97_stage_snapshot_before
            ==
            cell97_stage_snapshot_after
        ),

    "no_artifact_write":
        True,
}


cell_97_pass = all(
    cell97_gates.values()
)


cell97_status = (
    "NEXT_PILOT_QUESTION_DRAFT_CREATED_PENDING_MACHINE_VALIDATION"

    if (
        cell_97_pass
    )

    else
    "NEXT_PILOT_QUESTION_DRAFT_CREATION_FAILED"
)


# ============================================================
# 15) Diagnostics
# ============================================================

print(
    "=== NEXT PILOT — QUESTION DRAFT CREATION ==="
)


print(
    "\n=== WORKFLOW ==="
)

print(
    "Current stage:",
    "STAGE_02_CREATE_QUESTION_DRAFT"
)

print(
    "Next stage:",
    "STAGE_03_MACHINE_VALIDATE_QUESTION"
)


print(
    "\n=== PILOT ==="
)

print(
    "Pilot ID:",
    cell97_packet.get(
        "pilot_case_id"
    )
)

print(
    "Archetype:",
    cell97_packet.get(
        "question_archetype"
    )
)

print(
    "Evidence parts:",
    len(
        cell97_target_parts
    )
)

print(
    "Target tokens:",
    cell97_packet.get(
        "target_total_tokens"
    )
)


print(
    "\n=== QUESTION DRAFT ==="
)

print(
    CELL97_QUESTION_TEXT
)

print(
    "Question SHA256:",
    cell97_question_text_sha256
)

print(
    "Question Draft ID:",
    cell97_question_draft_id
)

print(
    "Draft record SHA256:",
    cell97_question_draft_record_sha256
)

print(
    "Draft lineage SHA256:",
    cell97_question_draft_lineage_sha256
)


print(
    "\n=== AUTHORING PREFLIGHT ==="
)

for gate_name, gate_value in (
    cell97_authoring_preflight.items()
):

    print(
        gate_name,
        ":",
        (
            "PASS"
            if gate_value
            else "FAIL"
        )
    )


print(
    "Source identity hits:",
    cell97_source_identity_hits
)

print(
    "Source-specific hits:",
    cell97_source_specific_hits
)

print(
    "Source-reference cue hits:",
    cell97_source_reference_cue_hits
)

print(
    "Answer leakage hits:",
    cell97_answer_leakage_hits
)


print(
    "\n=== DOWNSTREAM STATUS ==="
)

print(
    "Machine validation:",
    (
        cell97_question_draft_record.get(
            "machine_validation_status"
        )

        if (
            cell97_question_draft_record
            is not None
        )

        else None
    )
)

print(
    "Human adjudication:",
    (
        cell97_question_draft_record.get(
            "human_adjudication_status"
        )

        if (
            cell97_question_draft_record
            is not None
        )

        else None
    )
)

print(
    "Expected answer:",
    (
        cell97_question_draft_record.get(
            "expected_answer_status"
        )

        if (
            cell97_question_draft_record
            is not None
        )

        else None
    )
)

print(
    "Citation validation:",
    (
        cell97_question_draft_record.get(
            "citation_target_validation_status"
        )

        if (
            cell97_question_draft_record
            is not None
        )

        else None
    )
)

print(
    "Full 572 generation:",
    cell97_packet.get(
        "full_572_generation"
    )
)


print(
    "\n=== FAILURES ==="
)

print(
    "Prerequisite failures:",
    cell97_prerequisite_failures
)

print(
    "Workflow failures:",
    cell97_workflow_failures
)

print(
    "Packet failures:",
    cell97_packet_failures
)

print(
    "Evidence failures:",
    cell97_evidence_failures
)

print(
    "Authoring preflight failures:",
    cell97_preflight_failures
)

print(
    "Schema failures:",
    cell97_schema_failures
)

print(
    "ID failures:",
    cell97_id_failures
)


print(
    "\nSTATUS:",
    cell97_status
)


print(
    "\n=== CODE CELL 97 GATES ==="
)


for gate_name, gate_value in (
    cell97_gates.items()
):

    print(
        gate_name,
        ":",
        (
            "PASS"
            if gate_value
            else "FAIL"
        )
    )


print(
    "\nFAILED GATES:",
    [
        gate_name

        for gate_name, gate_value
        in cell97_gates.items()

        if not gate_value
    ]
)


print(
    "\nCODE CELL 97 GATE:",
    (
        "PASS"
        if cell_97_pass
        else "FAIL"
    )
)


assert cell_97_pass, (
    "Next Pilot Question Draft creation failed. "
    "Do not run Machine Validation. "
    "Repair this SAME Code Cell 97 using the printed diagnostics."
)
```

<h3>Machine Validation برای Question Draft دوم</h3>

<p>
Cell 97 Question Draft مربوط به Pilot دوم را با موفقیت ساخت و وضعیت آن اکنون
<code>DRAFT_CREATED_PENDING_VALIDATION</code>

</p>

<p>
در این مرحله همان Machine Validation فریز‌شده‌ی Pilot مرجع روی Question جدید اجرا می‌شود:
</p>

<ul>
<li>تمام Hard Gateهای <code>QUESTION_DRAFT_ACCEPTANCE_RUBRIC_V1</code>؛</li>
<li>تمام Semantic Gateها؛</li>
<li>قاعده‌ی <code>SOURCE_AGNOSTIC</code>؛</li>
<li>عدم Answer Leakage؛</li>
<li>عدم استفاده از External Knowledge؛</li>
<li>حفظ Internal Attribution و Authority؛</li>
<li>Groundability کامل در Target Evidence؛</li>
<li>Testable بودن تمام Required Dimensionهای همین Pilot؛</li>
<li>عدم ساخت Locator یا Metadata جدید.</li>
</ul>

<p>
برای این Pilot، سؤال از نوع <code>EVIDENCE_COMPLETENESS</code> است؛ بنابراین Machine Validation باید تأیید کند که Evidence انتخاب‌شده واقعاً تمام موضوعات لازم و تعداد کل مفاهیم را برای پاسخ کامل در اختیار دارد.
</p>

<p>
Machine Validation به‌تنهایی Question را Accept نمی‌کند. حتی در صورت PASS، مرحله‌ی بعد همچنان Human Adjudication اجباری خواهد بود.
</p>

<h4>معیار PASS</h4>

<ul>
<li>تمام Hard Gateها PASS باشند؛</li>
<li>تمام Semantic Gateها PASS باشند؛</li>
<li>تمام Required Dimensionها قابل ارزیابی باشند؛</li>
<li>Internal Provenance معتبر ولی از User-Facing Question مخفی باشد؛</li>
<li>Question Draft و Evidence و Queue هیچ Mutation نداشته باشند؛</li>
<li>Expected Answer و Citation Mapping همچنان Blocked بمانند؛</li>
<li>Full 572 generation همچنان Blocked بماند.</li>
</ul>

</div>

```python
# Code Cell 98

# ============================================================
# Stage 03 — Machine Validation of second Pilot Question.
#
# Reuses the frozen Question acceptance rubric and
# source-opacity policy from the completed reference Pilot.
#
# No human acceptance.
# No expected-answer generation.
# No citation mapping.
# No active-state mutation.
# No artifact write.
# ============================================================


# ============================================================
# 1) Prerequisites
# ============================================================

cell98_prerequisite_failures = []


if (
    globals().get(
        "cell_97_pass"
    )
    is not True
):

    cell98_prerequisite_failures.append(
        "CELL97_NOT_PASS"
    )


if (
    globals().get(
        "cell97_status"
    )
    !=
    "NEXT_PILOT_QUESTION_DRAFT_CREATED_PENDING_MACHINE_VALIDATION"
):

    cell98_prerequisite_failures.append(
        {
            "reason":
                "CELL97_STATUS_UNEXPECTED",

            "actual":
                globals().get(
                    "cell97_status"
                ),
        }
    )


CELL98_REQUIRED_RUNTIME_OBJECTS = [
    "cell97_question_draft_record",
    "cell97_question_draft_record_sha256",
    "cell97_question_draft_lineage",
    "cell97_question_draft_lineage_sha256",
    "cell96_next_pilot_standard_workflow_stage",
    "cell96_selected_task",
    "cell95_remaining_pilot_queue",
    "cell95_pilot_workflow_state",
    "PILOT_QA_WORKFLOW_CONTRACT_V1",
    "QUESTION_DRAFT_ACCEPTANCE_RUBRIC_V1",
    "QUESTION_USER_FACING_SOURCE_OPACITY_RULE_V1",
]


cell98_missing_runtime_objects = [
    name

    for name
    in CELL98_REQUIRED_RUNTIME_OBJECTS

    if (
        name
        not in
        globals()
    )
]


if (
    cell98_missing_runtime_objects
):

    cell98_prerequisite_failures.append(
        {
            "reason":
                "REQUIRED_RUNTIME_OBJECTS_MISSING",

            "objects":
                cell98_missing_runtime_objects,
        }
    )


CELL98_EXPECTED_CELL97 = {
    "question_text_sha256":
        "68af4d618d89d693db7c72ccd738086008170fe96e341c3c906bb4b3080525e3",

    "question_draft_id":
        "question-draft:ae250fc53d9a4c5c5691d9ca2d8a7429",

    "question_draft_record_sha256":
        "997f4855e160a78074086537d03f60dcb99131215855c4aa12a1fbcb5a1726ac",

    "question_draft_lineage_sha256":
        "01370df6b2af47991544c83210df010da48ba15946cf37c67f2ba85b2a79a453",
}


if (
    globals().get(
        "cell97_question_draft_record_sha256"
    )
    !=
    CELL98_EXPECTED_CELL97[
        "question_draft_record_sha256"
    ]
):

    cell98_prerequisite_failures.append(
        "CELL97_DRAFT_RECORD_SHA256_MISMATCH"
    )


if (
    globals().get(
        "cell97_question_draft_lineage_sha256"
    )
    !=
    CELL98_EXPECTED_CELL97[
        "question_draft_lineage_sha256"
    ]
):

    cell98_prerequisite_failures.append(
        "CELL97_DRAFT_LINEAGE_SHA256_MISMATCH"
    )


# ============================================================
# 2) Helper
# ============================================================

def cell98_sha(value):

    return hashlib.sha256(
        stable_json_key(
            value
        ).encode(
            "utf-8"
        )
    ).hexdigest()


# ============================================================
# 3) Validate frozen Workflow stage
# ============================================================

cell98_workflow_failures = []


cell98_lifecycle_stages = (
    PILOT_QA_WORKFLOW_CONTRACT_V1.get(
        "lifecycle_stages",
        [],
    )
)


if not (
    len(
        cell98_lifecycle_stages
    )
    >=
    4

    and

    cell98_lifecycle_stages[
        2
    ]
    ==
    "STAGE_03_MACHINE_VALIDATE_QUESTION"

    and

    cell98_lifecycle_stages[
        3
    ]
    ==
    "STAGE_04_HUMAN_ADJUDICATE_QUESTION"
):

    cell98_workflow_failures.append(
        {
            "reason":
                "FROZEN_WORKFLOW_STAGE_SEQUENCE_UNEXPECTED",

            "actual":
                cell98_lifecycle_stages,
        }
    )


if (
    PILOT_QA_WORKFLOW_CONTRACT_V1.get(
        "machine_validation_requirements",
        {},
    ).get(
        "question"
    )
    !=
    "FROZEN_ACCEPTANCE_RUBRIC_PLUS_SOURCE_OPACITY"
):

    cell98_workflow_failures.append(
        "QUESTION_MACHINE_VALIDATION_CONTRACT_UNEXPECTED"
    )


if (
    PILOT_QA_WORKFLOW_CONTRACT_V1.get(
        "automatic_acceptance"
    )
    is not False
):

    cell98_workflow_failures.append(
        "AUTOMATIC_ACCEPTANCE_NOT_FALSE"
    )


# ============================================================
# 4) Freeze protected state BEFORE validation
# ============================================================

CELL98_ACTIVE_STATE_OBJECT_NAMES = [
    "golden_case_records",
    "case_groundability_precheck",
    "case_task_design",
    "task_draft_plan_rows",
    "target_evidence_records",
    "pilot_selection_rows",
    "golden_case_records_by_id",
    "task_draft_plans",
    "target_evidence_by_case_id",
    "target_evidence_by_task_draft_id",
]


cell98_active_snapshot_before = (
    cell98_sha(
        {
            name:
                globals()[name]

            for name
            in CELL98_ACTIVE_STATE_OBJECT_NAMES

            if (
                name
                in
                globals()
            )
        }
    )
)


cell98_pilot_snapshot_before = (
    cell98_sha(
        {
            "question_drafting_pilot_packets":
                question_drafting_pilot_packets,

            "question_drafting_pilot_by_id":
                question_drafting_pilot_by_id,
        }
    )
)


cell98_queue_snapshot_before = (
    cell98_sha(
        cell95_remaining_pilot_queue
    )
)


cell98_workflow_state_snapshot_before = (
    cell98_sha(
        cell95_pilot_workflow_state
    )
)


cell98_staging_snapshot_before = (
    cell98_sha(
        cell96_next_pilot_standard_workflow_stage
    )
)


cell98_question_snapshot_before = (
    cell98_sha(
        cell97_question_draft_record
    )
)


cell98_lineage_snapshot_before = (
    cell98_sha(
        cell97_question_draft_lineage
    )
)


# ============================================================
# 5) Resolve Question / Target / Rubric
# ============================================================

cell98_question_record = (
    cell97_question_draft_record
)


cell98_question_text = (
    cell98_question_record[
        "question_text"
    ]
)


cell98_target_parts = (
    cell96_next_pilot_standard_workflow_stage[
        "target_text_parts"
    ]
)


cell98_joined_evidence = "\n".join(
    part[
        "text"
    ]

    for part
    in cell98_target_parts
)


cell98_rubric = (
    QUESTION_DRAFT_ACCEPTANCE_RUBRIC_V1
)


cell98_hard_gate_contract = (
    cell98_rubric.get(
        "hard_gates",
        {},
    )
)


cell98_semantic_gate_contract = (
    cell98_rubric.get(
        "semantic_gates",
        {},
    )
)


cell98_resolution_failures = []


if (
    cell98_question_record.get(
        "question_text_sha256"
    )
    !=
    CELL98_EXPECTED_CELL97[
        "question_text_sha256"
    ]
):

    cell98_resolution_failures.append(
        "QUESTION_TEXT_SHA256_UNEXPECTED"
    )


if (
    cell98_question_record.get(
        "question_draft_id"
    )
    !=
    CELL98_EXPECTED_CELL97[
        "question_draft_id"
    ]
):

    cell98_resolution_failures.append(
        "QUESTION_DRAFT_ID_UNEXPECTED"
    )


if (
    cell98_question_record.get(
        "machine_validation_status"
    )
    !=
    "NOT_STARTED"
):

    cell98_resolution_failures.append(
        "MACHINE_VALIDATION_STATUS_NOT_NOT_STARTED"
    )


if (
    cell97_question_draft_lineage.get(
        "next_action"
    )
    !=
    "STAGE_03_MACHINE_VALIDATE_QUESTION"
):

    cell98_resolution_failures.append(
        "QUESTION_DRAFT_LINEAGE_NEXT_ACTION_UNEXPECTED"
    )


# ============================================================
# 6) Validate internal provenance
# ============================================================

cell98_internal_attribution_names = set()

cell98_valid_direct_author_parts = []


for part in (
    cell98_target_parts
):

    attribution = (
        part.get(
            "attribution"
        )
        or
        {}
    )


    for field_name in [
        "author",
        "credited_author_name",
        "knowledge_owner",
        "reference_expert_name",
        "editorial_supervisor_name",
    ]:

        value = (
            attribution.get(
                field_name
            )
        )


        if (
            isinstance(
                value,
                str,
            )

            and

            value.strip()
        ):

            cell98_internal_attribution_names.add(
                value.strip()
            )


    if (
        attribution.get(
            "attribution_status"
        )
        ==
        "policy_validated"

        and

        attribution.get(
            "content_authority_type"
        )
        ==
        "direct_author"
    ):

        cell98_valid_direct_author_parts.append(
            part
        )


cell98_internal_provenance_failures = []


if (
    len(
        cell98_valid_direct_author_parts
    )
    ==
    0
):

    cell98_internal_provenance_failures.append(
        "NO_POLICY_VALIDATED_DIRECT_AUTHOR_EVIDENCE"
    )


if (
    cell96_next_pilot_standard_workflow_stage.get(
        "authority_expectation"
    )
    !=
    "PRIMARY_DIRECT_AUTHOR"
):

    cell98_internal_provenance_failures.append(
        {
            "reason":
                "AUTHORITY_EXPECTATION_UNEXPECTED",

            "actual":
                cell96_next_pilot_standard_workflow_stage.get(
                    "authority_expectation"
                ),
        }
    )


# ============================================================
# 7) Source-opacity product gates
# ============================================================

CELL98_SOURCE_REFERENCE_CUES = [
    "بر اساس توضیح",
    "بر اساس نظر",
    "به گفته",
    "طبق گفته",
    "طبق نظر",
    "از دیدگاه",
    "از نظر دکتر",
    "از نظر نویسنده",
    "در این مقاله",
    "در این منبع",
    "منبع",
    "نویسنده",
]


cell98_source_identity_hits = [
    name

    for name
    in sorted(
        cell98_internal_attribution_names
    )

    if (
        name
        in
        cell98_question_text
    )
]


cell98_source_reference_cue_hits = [
    cue

    for cue
    in CELL98_SOURCE_REFERENCE_CUES

    if (
        cue
        in
        cell98_question_text
    )
]


cell98_source_opacity_gate_results = {
    "source_identity_not_exposed":
        (
            len(
                cell98_source_identity_hits
            )
            ==
            0
        ),

    "source_reference_cue_not_exposed":
        (
            len(
                cell98_source_reference_cue_hits
            )
            ==
            0
        ),

    "user_facing_question_policy_is_source_agnostic":
        (
            QUESTION_USER_FACING_SOURCE_OPACITY_RULE_V1.get(
                "user_facing_question_policy"
            )
            ==
            "SOURCE_AGNOSTIC"
        ),

    "internal_provenance_is_required":
        (
            QUESTION_USER_FACING_SOURCE_OPACITY_RULE_V1.get(
                "internal_provenance_retention"
            )
            ==
            "REQUIRED"
        ),

    "internal_authority_validation_is_required":
        (
            QUESTION_USER_FACING_SOURCE_OPACITY_RULE_V1.get(
                "internal_source_authority_validation"
            )
            ==
            "REQUIRED"
        ),

    "internal_attribution_validation_is_required":
        (
            QUESTION_USER_FACING_SOURCE_OPACITY_RULE_V1.get(
                "internal_attribution_validation"
            )
            ==
            "REQUIRED"
        ),
}


cell98_failed_source_opacity_gates = [
    gate_name

    for gate_name, gate_value
    in cell98_source_opacity_gate_results.items()

    if not gate_value
]


# ============================================================
# 8) Evidence-completeness anchors
# ============================================================

CELL98_REQUIRED_ANSWER_EVIDENCE_ANCHORS = [
    "خرج",
    "پس‌انداز",
    "شوک‌های درآمدی",
    "هزین‌های",
    "اعتبار",
    "بدهی",
    "سرمایه‌گذاری",
    "مدیریت ریسک",
    "تصمیم‌گیری مالی",
    "چهارده مفهوم سواد مالی",
]


cell98_missing_evidence_anchors = [
    anchor

    for anchor
    in CELL98_REQUIRED_ANSWER_EVIDENCE_ANCHORS

    if (
        anchor
        not in
        cell98_joined_evidence
    )
]


CELL98_ANSWER_LEAKAGE_ANCHORS = [
    "خرج",
    "پس‌انداز",
    "شوک‌های درآمدی",
    "هزین‌های",
    "اعتبار",
    "بدهی",
    "سرمایه‌گذاری",
    "مدیریت ریسک",
    "تصمیم‌گیری مالی",
    "چهارده",
    "۱۴",
]


cell98_answer_leakage_hits = [
    anchor

    for anchor
    in CELL98_ANSWER_LEAKAGE_ANCHORS

    if (
        anchor
        in
        cell98_question_text
    )
]


CELL98_INTERNAL_METADATA_MARKERS = [
    "golden-case:",
    "task-draft:",
    "target-evidence:",
    "question-pilot:",
    "question-draft:",
    "sha256",
    "candidate_identity",
    "document_id",
    "source_id",
]


cell98_internal_metadata_hits = [
    marker

    for marker
    in CELL98_INTERNAL_METADATA_MARKERS

    if (
        marker
        in
        cell98_question_text
    )
]


CELL98_LOCATOR_MARKERS = [
    "http://",
    "https://",
    "www.",
    "صفحه ",
    "خط ",
    "فایل ",
]


cell98_locator_hits = [
    marker

    for marker
    in CELL98_LOCATOR_MARKERS

    if (
        marker
        in
        cell98_question_text
    )
]


# ============================================================
# 9) Frozen Hard Gates
# ============================================================

cell98_hard_gate_results = {
    "nonempty_question_text":
        (
            isinstance(
                cell98_question_text,
                str,
            )

            and

            len(
                cell98_question_text.strip()
            )
            >
            0
        ),

    "natural_user_question":
        (
            isinstance(
                cell98_question_text,
                str,
            )

            and

            len(
                cell98_question_text.strip()
            )
            >=
            20

            and

            cell98_question_text.strip().endswith(
                "؟"
            )

            and

            "\n"
            not in
            cell98_question_text.strip()
        ),

    "external_knowledge_dependency":
        (
            len(
                cell98_missing_evidence_anchors
            )
            ==
            0
        ),

    "answer_leakage_in_question":
        (
            len(
                cell98_answer_leakage_hits
            )
            ==
            0
        ),

    "internal_document_id_leakage":
        (
            not any(
                marker
                in
                cell98_question_text

                for marker
                in [
                    "document_id",
                    "source_id",
                    "golden-case:",
                    "task-draft:",
                    "target-evidence:",
                    "question-pilot:",
                ]
            )
        ),

    "internal_pipeline_metadata_leakage":
        (
            len(
                cell98_internal_metadata_hits
            )
            ==
            0
        ),

    "unknown_speaker_inference":
        (
            "گوینده"
            not in
            cell98_question_text

            and

            "speaker"
            not in
            cell98_question_text.lower()
        ),

    "unsupported_attribution":
        (
            len(
                cell98_internal_attribution_names
            )
            >
            0

            and

            len(
                cell98_valid_direct_author_parts
            )
            >
            0
        ),

    "unsupported_authority_promotion":
        (
            cell96_next_pilot_standard_workflow_stage.get(
                "authority_expectation"
            )
            ==
            "PRIMARY_DIRECT_AUTHOR"

            and

            len(
                cell98_valid_direct_author_parts
            )
            >=
            1
        ),

    "locator_fabrication":
        (
            len(
                cell98_locator_hits
            )
            ==
            0
        ),

    "question_must_be_groundable_in_target_text_parts":
        (
            len(
                cell98_target_parts
            )
            >=
            1

            and

            len(
                cell98_missing_evidence_anchors
            )
            ==
            0
        ),

    "target_evidence_must_already_be_validated":
        (
            cell96_selected_task.get(
                "target_evidence_status"
            )
            ==
            "SELECTED_AND_VALIDATED"
        ),
}


cell98_failed_hard_gates = [
    gate_name

    for gate_name, gate_value
    in cell98_hard_gate_results.items()

    if not gate_value
]


# ============================================================
# 10) Required-dimension testability
# ============================================================

cell98_required_dimensions = (
    cell96_next_pilot_standard_workflow_stage[
        "required_dimensions"
    ]
)


CELL98_DIMENSION_TESTABILITY = {
    "retrieval_hit":
        (
            "The question requires explicit factual evidence "
            "from the validated target."
        ),

    "rank":
        (
            "The validated target can be evaluated for ranking "
            "against competing retrieval candidates."
        ),

    "evidence_completeness":
        (
            "The question explicitly requires the complete set of "
            "stated financial-literacy topics plus the stated total."
        ),

    "attribution_correctness":
        (
            "Retrieved evidence attribution can be validated "
            "internally without exposing source identity to the user."
        ),

    "citation_locator_correctness":
        (
            "Retrieved evidence can be checked against the exact "
            "validated internal locator."
        ),

    "unnecessary_context":
        (
            "The question targets only the enumerated concepts and "
            "their total, allowing excess retrieved context to be measured."
        ),

    "source_authority_correctness":
        (
            "Retrieved evidence authority can be validated internally "
            "against PRIMARY_DIRECT_AUTHOR."
        ),
}


cell98_uncovered_required_dimensions = [
    dimension

    for dimension
    in cell98_required_dimensions

    if (
        dimension
        not in
        CELL98_DIMENSION_TESTABILITY
    )
]


# ============================================================
# 11) Frozen Semantic Gates
# ============================================================

CELL98_SEMANTIC_GATE_JUDGMENTS = {
    "task_family_alignment": {
        "status":
            (
                "PASS"

                if (
                    cell96_next_pilot_standard_workflow_stage.get(
                        "task_family"
                    )
                    ==
                    "LONG_DIRECT_PRIMARY_SOURCE_RETRIEVAL"
                )

                else
                "FAIL"
            ),

        "reason":
            (
                "The question requests substantive factual content "
                "from the validated long direct-primary target."
            ),
    },

    "question_archetype_alignment": {
        "status":
            (
                "PASS"

                if (
                    cell96_next_pilot_standard_workflow_stage.get(
                        "question_archetype"
                    )
                    ==
                    "EVIDENCE_COMPLETENESS"
                )

                else
                "FAIL"
            ),

        "reason":
            (
                "The question requires both the explicitly stated "
                "financial-literacy topics and the stated total count."
            ),
    },

    "evidence_scope_alignment": {
        "status":
            (
                "PASS"

                if (
                    cell96_next_pilot_standard_workflow_stage.get(
                        "evidence_scope"
                    )
                    ==
                    "SINGLE_LONG_SOURCE_DOCUMENT"
                )

                else
                "FAIL"
            ),

        "reason":
            (
                "All required answer components are contained in "
                "the single validated target evidence window."
            ),
    },

    "authority_expectation_alignment": {
        "status":
            (
                "PASS"

                if (
                    len(
                        cell98_internal_provenance_failures
                    )
                    ==
                    0
                )

                else
                "FAIL"
            ),

        "reason":
            (
                "PRIMARY_DIRECT_AUTHOR is validated internally "
                "without exposing the source or author to the user."
            ),
    },

    "context_expectation_alignment": {
        "status":
            (
                "PASS"

                if (
                    cell96_next_pilot_standard_workflow_stage.get(
                        "context_expectation"
                    )
                    ==
                    "SUFFICIENT_WITHOUT_EXCESS"
                )

                else
                "FAIL"
            ),

        "reason":
            (
                "The question can be answered from the selected "
                "minimal evidence window without requiring extra context."
            ),
    },

    "required_dimensions_are_testable_by_question": {
        "status":
            (
                "PASS"

                if (
                    len(
                        cell98_uncovered_required_dimensions
                    )
                    ==
                    0
                )

                else
                "FAIL"
            ),

        "reason":
            (
                "Every required Pilot dimension has an explicit "
                "machine-evaluation interpretation."
            ),
    },

    "question_is_not_trivial_metadata_repetition": {
        "status":
            (
                "PASS"

                if (
                    len(
                        cell98_internal_metadata_hits
                    )
                    ==
                    0
                )

                else
                "FAIL"
            ),

        "reason":
            (
                "The question requests substantive financial-literacy "
                "content rather than source or pipeline metadata."
            ),
    },

    "question_is_answerable_without_external_context": {
        "status":
            (
                "PASS"

                if (
                    len(
                        cell98_missing_evidence_anchors
                    )
                    ==
                    0
                )

                else
                "FAIL"
            ),

        "reason":
            (
                "The validated Target Evidence contains all required "
                "answer components and the explicit total."
            ),
    },
}


cell98_semantic_gate_results = {
    gate_name:
        (
            judgment[
                "status"
            ]
            ==
            "PASS"
        )

    for gate_name, judgment
    in CELL98_SEMANTIC_GATE_JUDGMENTS.items()
}


cell98_failed_semantic_gates = [
    gate_name

    for gate_name, gate_value
    in cell98_semantic_gate_results.items()

    if not gate_value
]


# ============================================================
# 12) Exact Rubric contract coverage
# ============================================================

cell98_contract_failures = []


if (
    set(
        cell98_hard_gate_results.keys()
    )
    !=
    set(
        cell98_hard_gate_contract.keys()
    )
):

    cell98_contract_failures.append(
        {
            "reason":
                "HARD_GATE_COVERAGE_NOT_EXACT",

            "expected":
                sorted(
                    cell98_hard_gate_contract.keys()
                ),

            "actual":
                sorted(
                    cell98_hard_gate_results.keys()
                ),
        }
    )


if (
    set(
        cell98_semantic_gate_results.keys()
    )
    !=
    set(
        cell98_semantic_gate_contract.keys()
    )
):

    cell98_contract_failures.append(
        {
            "reason":
                "SEMANTIC_GATE_COVERAGE_NOT_EXACT",

            "expected":
                sorted(
                    cell98_semantic_gate_contract.keys()
                ),

            "actual":
                sorted(
                    cell98_semantic_gate_results.keys()
                ),
        }
    )


# ============================================================
# 13) Machine decision
# ============================================================

cell98_machine_validation_decision = (
    "PASS"

    if (
        len(
            cell98_prerequisite_failures
        )
        ==
        0

        and

        len(
            cell98_workflow_failures
        )
        ==
        0

        and

        len(
            cell98_resolution_failures
        )
        ==
        0

        and

        len(
            cell98_internal_provenance_failures
        )
        ==
        0

        and

        len(
            cell98_failed_source_opacity_gates
        )
        ==
        0

        and

        len(
            cell98_failed_hard_gates
        )
        ==
        0

        and

        len(
            cell98_failed_semantic_gates
        )
        ==
        0

        and

        len(
            cell98_uncovered_required_dimensions
        )
        ==
        0

        and

        len(
            cell98_contract_failures
        )
        ==
        0
    )

    else
    "FAIL"
)


# ============================================================
# 14) Materialize Machine Validation record
# ============================================================

cell98_source_opacity_rule_sha256 = (
    cell98_sha(
        QUESTION_USER_FACING_SOURCE_OPACITY_RULE_V1
    )
)


cell98_machine_validation_record = {
    "validation_version":
        "question_draft_machine_validation_v2_source_opaque",

    "workflow_stage":
        "STAGE_03_MACHINE_VALIDATE_QUESTION",

    "question_draft_id":
        cell98_question_record[
            "question_draft_id"
        ],

    "pilot_case_id":
        cell98_question_record[
            "pilot_case_id"
        ],

    "question_text":
        cell98_question_text,

    "question_text_sha256":
        cell98_question_record[
            "question_text_sha256"
        ],

    "question_draft_record_sha256":
        cell97_question_draft_record_sha256,

    "question_draft_lineage_sha256":
        cell97_question_draft_lineage_sha256,

    "rubric_version":
        cell98_rubric[
            "version"
        ],

    "source_opacity_rule_version":
        QUESTION_USER_FACING_SOURCE_OPACITY_RULE_V1[
            "version"
        ],

    "source_opacity_rule_sha256":
        cell98_source_opacity_rule_sha256,

    "source_opacity_gate_results":
        dict(
            cell98_source_opacity_gate_results
        ),

    "hard_gate_results":
        dict(
            cell98_hard_gate_results
        ),

    "semantic_gate_results":
        dict(
            cell98_semantic_gate_results
        ),

    "semantic_gate_judgments":
        dict(
            CELL98_SEMANTIC_GATE_JUDGMENTS
        ),

    "required_dimension_testability":
        {
            dimension:
                CELL98_DIMENSION_TESTABILITY[
                    dimension
                ]

            for dimension
            in cell98_required_dimensions
        },

    "internal_provenance_status":
        "VALIDATED_AND_HIDDEN_FROM_USER",

    "machine_validation_decision":
        cell98_machine_validation_decision,

    "human_adjudication_required":
        True,

    "human_adjudication_status":
        "NOT_STARTED",

    "automatic_acceptance":
        False,

    "expected_answer_status":
        "BLOCKED_PENDING_QUESTION_ACCEPTANCE",

    "citation_target_validation_status":
        "BLOCKED_PENDING_QUESTION_ACCEPTANCE",

    "full_572_generation":
        PILOT_QA_WORKFLOW_CONTRACT_V1[
            "full_population_generation"
        ],

    "next_action":
        "STAGE_04_HUMAN_ADJUDICATE_QUESTION",
}


cell98_machine_validation_record_sha256 = (
    cell98_sha(
        cell98_machine_validation_record
    )
)


# ============================================================
# 15) Freeze protected state AFTER validation
# ============================================================

cell98_active_snapshot_after = (
    cell98_sha(
        {
            name:
                globals()[name]

            for name
            in CELL98_ACTIVE_STATE_OBJECT_NAMES

            if (
                name
                in
                globals()
            )
        }
    )
)


cell98_pilot_snapshot_after = (
    cell98_sha(
        {
            "question_drafting_pilot_packets":
                question_drafting_pilot_packets,

            "question_drafting_pilot_by_id":
                question_drafting_pilot_by_id,
        }
    )
)


cell98_queue_snapshot_after = (
    cell98_sha(
        cell95_remaining_pilot_queue
    )
)


cell98_workflow_state_snapshot_after = (
    cell98_sha(
        cell95_pilot_workflow_state
    )
)


cell98_staging_snapshot_after = (
    cell98_sha(
        cell96_next_pilot_standard_workflow_stage
    )
)


cell98_question_snapshot_after = (
    cell98_sha(
        cell97_question_draft_record
    )
)


cell98_lineage_snapshot_after = (
    cell98_sha(
        cell97_question_draft_lineage
    )
)


# ============================================================
# 16) Final gates
# ============================================================

cell98_gates = {
    "cell97_is_pass":
        (
            globals().get(
                "cell_97_pass"
            )
            is True
        ),

    "workflow_stage_03_is_exact":
        (
            len(
                cell98_workflow_failures
            )
            ==
            0
        ),

    "question_draft_handoff_is_exact":
        (
            len(
                cell98_resolution_failures
            )
            ==
            0
        ),

    "source_opacity_rule_is_enforced":
        (
            len(
                cell98_failed_source_opacity_gates
            )
            ==
            0
        ),

    "internal_provenance_remains_valid":
        (
            len(
                cell98_internal_provenance_failures
            )
            ==
            0
        ),

    "all_frozen_hard_gates_pass":
        (
            len(
                cell98_failed_hard_gates
            )
            ==
            0
        ),

    "all_frozen_semantic_gates_pass":
        (
            len(
                cell98_failed_semantic_gates
            )
            ==
            0
        ),

    "all_required_dimensions_are_testable":
        (
            len(
                cell98_uncovered_required_dimensions
            )
            ==
            0
        ),

    "rubric_contract_coverage_is_exact":
        (
            len(
                cell98_contract_failures
            )
            ==
            0
        ),

    "machine_validation_decision_is_pass":
        (
            cell98_machine_validation_decision
            ==
            "PASS"
        ),

    "human_acceptance_is_still_required":
        (
            cell98_machine_validation_record[
                "human_adjudication_required"
            ]
            is True

            and

            cell98_machine_validation_record[
                "human_adjudication_status"
            ]
            ==
            "NOT_STARTED"

            and

            cell98_machine_validation_record[
                "automatic_acceptance"
            ]
            is False
        ),

    "expected_answer_remains_blocked":
        (
            cell98_machine_validation_record[
                "expected_answer_status"
            ]
            ==
            "BLOCKED_PENDING_QUESTION_ACCEPTANCE"
        ),

    "citation_validation_remains_blocked":
        (
            cell98_machine_validation_record[
                "citation_target_validation_status"
            ]
            ==
            "BLOCKED_PENDING_QUESTION_ACCEPTANCE"
        ),

    "full_572_generation_remains_blocked":
        (
            cell98_machine_validation_record[
                "full_572_generation"
            ]
            ==
            "BLOCKED_PENDING_FULL_PILOT_WORKFLOW_VALIDATION"
        ),

    "question_record_remains_byte_stable":
        (
            cell98_question_snapshot_before
            ==
            cell98_question_snapshot_after
        ),

    "question_lineage_remains_byte_stable":
        (
            cell98_lineage_snapshot_before
            ==
            cell98_lineage_snapshot_after
        ),

    "staging_packet_remains_byte_stable":
        (
            cell98_staging_snapshot_before
            ==
            cell98_staging_snapshot_after
        ),

    "active_state_remains_byte_stable":
        (
            cell98_active_snapshot_before
            ==
            cell98_active_snapshot_after
        ),

    "pilot_runtime_remains_byte_stable":
        (
            cell98_pilot_snapshot_before
            ==
            cell98_pilot_snapshot_after
        ),

    "remaining_queue_remains_byte_stable":
        (
            cell98_queue_snapshot_before
            ==
            cell98_queue_snapshot_after
        ),

    "workflow_state_remains_byte_stable":
        (
            cell98_workflow_state_snapshot_before
            ==
            cell98_workflow_state_snapshot_after
        ),

    "no_artifact_write":
        True,
}


cell_98_pass = all(
    cell98_gates.values()
)


cell98_status = (
    "NEXT_PILOT_QUESTION_MACHINE_PASS_PENDING_HUMAN_ADJUDICATION"

    if (
        cell_98_pass
    )

    else
    "NEXT_PILOT_QUESTION_MACHINE_VALIDATION_FAILED"
)


# ============================================================
# 17) Diagnostics
# ============================================================

print(
    "=== NEXT PILOT — QUESTION MACHINE VALIDATION ==="
)


print(
    "\n=== WORKFLOW ==="
)

print(
    "Current stage:",
    "STAGE_03_MACHINE_VALIDATE_QUESTION"
)

print(
    "Next stage:",
    "STAGE_04_HUMAN_ADJUDICATE_QUESTION"
)


print(
    "\n=== QUESTION ==="
)

print(
    cell98_question_text
)


print(
    "\n=== SOURCE OPACITY GATES ==="
)


for gate_name, gate_value in (
    cell98_source_opacity_gate_results.items()
):

    print(
        gate_name,
        ":",
        (
            "PASS"
            if gate_value
            else "FAIL"
        )
    )


print(
    "Identity hits:",
    cell98_source_identity_hits
)

print(
    "Source-reference cue hits:",
    cell98_source_reference_cue_hits
)

print(
    "Failed source-opacity gates:",
    cell98_failed_source_opacity_gates
)


print(
    "\n=== INTERNAL PROVENANCE ==="
)

print(
    "Attribution names:",
    sorted(
        cell98_internal_attribution_names
    )
)

print(
    "Validated direct-author parts:",
    len(
        cell98_valid_direct_author_parts
    )
)

print(
    "Failures:",
    cell98_internal_provenance_failures
)


print(
    "\n=== EVIDENCE COMPLETENESS ==="
)

print(
    "Required answer evidence anchors:",
    CELL98_REQUIRED_ANSWER_EVIDENCE_ANCHORS
)

print(
    "Missing evidence anchors:",
    cell98_missing_evidence_anchors
)

print(
    "Answer leakage hits:",
    cell98_answer_leakage_hits
)


print(
    "\n=== HARD GATES ==="
)


for gate_name in (
    cell98_hard_gate_contract.keys()
):

    print(
        gate_name,
        ":",
        (
            "PASS"
            if (
                cell98_hard_gate_results[
                    gate_name
                ]
            )
            else
            "FAIL"
        )
    )


print(
    "Failed hard gates:",
    cell98_failed_hard_gates
)


print(
    "\n=== SEMANTIC GATES ==="
)


for gate_name in (
    cell98_semantic_gate_contract.keys()
):

    judgment = (
        CELL98_SEMANTIC_GATE_JUDGMENTS[
            gate_name
        ]
    )


    print(
        "\n-",
        gate_name,
        ":",
        judgment[
            "status"
        ]
    )

    print(
        "  Reason:",
        judgment[
            "reason"
        ]
    )


print(
    "\nFailed semantic gates:",
    cell98_failed_semantic_gates
)


print(
    "\n=== REQUIRED DIMENSION TESTABILITY ==="
)


for dimension in (
    cell98_required_dimensions
):

    print(
        "-",
        dimension,
        ":",
        CELL98_DIMENSION_TESTABILITY.get(
            dimension
        )
    )


print(
    "Uncovered dimensions:",
    cell98_uncovered_required_dimensions
)


print(
    "\n=== MACHINE DECISION ==="
)

print(
    "Decision:",
    cell98_machine_validation_decision
)

print(
    "Validation record SHA256:",
    cell98_machine_validation_record_sha256
)


print(
    "\n=== DOWNSTREAM BOUNDARY ==="
)

print(
    "Human adjudication:",
    cell98_machine_validation_record[
        "human_adjudication_status"
    ]
)

print(
    "Automatic acceptance:",
    cell98_machine_validation_record[
        "automatic_acceptance"
    ]
)

print(
    "Expected answer:",
    cell98_machine_validation_record[
        "expected_answer_status"
    ]
)

print(
    "Citation validation:",
    cell98_machine_validation_record[
        "citation_target_validation_status"
    ]
)

print(
    "Full 572 generation:",
    cell98_machine_validation_record[
        "full_572_generation"
    ]
)


print(
    "\n=== FAILURES ==="
)

print(
    "Prerequisite failures:",
    cell98_prerequisite_failures
)

print(
    "Workflow failures:",
    cell98_workflow_failures
)

print(
    "Resolution failures:",
    cell98_resolution_failures
)

print(
    "Internal provenance failures:",
    cell98_internal_provenance_failures
)

print(
    "Source-opacity failures:",
    cell98_failed_source_opacity_gates
)

print(
    "Hard-gate failures:",
    cell98_failed_hard_gates
)

print(
    "Semantic-gate failures:",
    cell98_failed_semantic_gates
)

print(
    "Contract failures:",
    cell98_contract_failures
)


print(
    "\nSTATUS:",
    cell98_status
)


print(
    "\n=== CODE CELL 98 GATES ==="
)


for gate_name, gate_value in (
    cell98_gates.items()
):

    print(
        gate_name,
        ":",
        (
            "PASS"
            if gate_value
            else "FAIL"
        )
    )


print(
    "\nFAILED GATES:",
    [
        gate_name

        for gate_name, gate_value
        in cell98_gates.items()

        if not gate_value
    ]
)


print(
    "\nCODE CELL 98 GATE:",
    (
        "PASS"
        if cell_98_pass
        else "FAIL"
    )
)


assert cell_98_pass, (
    "Second Pilot Question failed Machine Validation. "
    "Do not perform Human Adjudication or generate an Expected Answer. "
    "Repair this SAME Code Cell 98 using the printed diagnostics."
)
```

```python
# Code Cell 102

# ============================================================
# Read-only upstream inspection for the same Website source.
#
# No correction.
# No artifact write.
# No expected-answer generation.
# ============================================================


# ============================================================
# 1) Prerequisites
# ============================================================

cell102_prerequisite_failures = []


if (
    globals().get(
        "cell_101_pass"
    )
    is not True
):

    cell102_prerequisite_failures.append(
        "CELL101_NOT_PASS"
    )


if (
    globals().get(
        "cell101_status"
    )
    !=
    "SAME_SOURCE_EVIDENCE_INSPECTED_PENDING_INTERPRETATION"
):

    cell102_prerequisite_failures.append(
        {
            "reason":
                "CELL101_STATUS_UNEXPECTED",

            "actual":
                globals().get(
                    "cell101_status"
                ),
        }
    )


CELL102_REQUIRED_RUNTIME_OBJECTS = [
    "cell101_source_record",
    "cell101_canonical_url",
    "cell101_source_token",
    "cell101_accepted_corpus_path",
    "cell101_verification_result",
    "cell100_human_adjudication_record",
]


cell102_missing_runtime_objects = [
    name

    for name
    in CELL102_REQUIRED_RUNTIME_OBJECTS

    if (
        name
        not in
        globals()
    )
]


if (
    cell102_missing_runtime_objects
):

    cell102_prerequisite_failures.append(
        {
            "reason":
                "REQUIRED_RUNTIME_OBJECTS_MISSING",

            "objects":
                cell102_missing_runtime_objects,
        }
    )


# ============================================================
# 2) Helpers
# ============================================================

def cell102_sha(value):

    return hashlib.sha256(
        stable_json_key(
            value
        ).encode(
            "utf-8"
        )
    ).hexdigest()


def cell102_walk_strings(value, path=()):

    if isinstance(
        value,
        dict,
    ):

        for key, child in value.items():

            yield from cell102_walk_strings(
                child,
                path + (str(key),),
            )

    elif isinstance(
        value,
        list,
    ):

        for index, child in enumerate(
            value
        ):

            yield from cell102_walk_strings(
                child,
                path + (str(index),),
            )

    elif isinstance(
        value,
        str,
    ):

        yield path, value


# ============================================================
# 3) Freeze protected state
# ============================================================

cell102_source_record_snapshot_before = (
    cell102_sha(
        cell101_source_record
    )
)


cell102_question_snapshot_before = (
    cell102_sha(
        cell97_question_draft_record
    )
)


cell102_adjudication_snapshot_before = (
    cell102_sha(
        cell100_human_adjudication_record
    )
)


cell102_pilot_runtime_snapshot_before = (
    cell102_sha(
        {
            "question_drafting_pilot_packets":
                question_drafting_pilot_packets,

            "question_drafting_pilot_by_id":
                question_drafting_pilot_by_id,
        }
    )
)


# ============================================================
# 4) Resolve repository root from accepted corpus
# ============================================================

cell102_repo_root = None


for parent in [
    cell101_accepted_corpus_path.parent,
    *cell101_accepted_corpus_path.parents,
]:

    if (
        (
            parent
            /
            "data"
            /
            "phase2"
        ).is_dir()
    ):

        cell102_repo_root = (
            parent
        )


if (
    cell102_repo_root
    is None
):

    cell102_prerequisite_failures.append(
        "REPOSITORY_ROOT_NOT_RESOLVED"
    )


# ============================================================
# 5) Discover lineage/path references from accepted record
# ============================================================

cell102_record_string_fields = list(
    cell102_walk_strings(
        cell101_source_record
    )
)


cell102_lineage_candidates = []


for path, value in (
    cell102_record_string_fields
):

    normalized = (
        value.strip()
    )


    if not normalized:

        continue


    looks_relevant = (
        cell101_source_token
        in
        normalized

        or

        "raw"
        in
        ".".join(
            path
        ).lower()

        or

        "source"
        in
        ".".join(
            path
        ).lower()

        or

        "interim"
        in
        ".".join(
            path
        ).lower()

        or

        normalized.endswith(
            (
                ".html",
                ".htm",
                ".json",
                ".jsonl",
                ".md",
                ".txt",
            )
        )
    )


    if (
        looks_relevant
    ):

        cell102_lineage_candidates.append(
            {
                "field_path":
                    ".".join(
                        path
                    ),

                "value":
                    normalized,
            }
        )


# ============================================================
# 6) Resolve directly referenced local files
# ============================================================

cell102_candidate_files = []


if (
    cell102_repo_root
    is not None
):

    for item in (
        cell102_lineage_candidates
    ):

        value = (
            item[
                "value"
            ]
        )


        possible_paths = [
            Path(
                value
            ),

            cell102_repo_root
            /
            value,
        ]


        for candidate in (
            possible_paths
        ):

            try:

                resolved = (
                    candidate.expanduser().resolve()
                )

            except Exception:

                continue


            if (
                resolved.is_file()

                and

                resolved
                !=
                cell101_accepted_corpus_path
            ):

                cell102_candidate_files.append(
                    resolved
                )


# ============================================================
# 7) Discover filenames containing exact source token
# ============================================================

cell102_filename_matches = []


if (
    cell102_repo_root
    is not None
):

    cell102_phase2_root = (
        cell102_repo_root
        /
        "data"
        /
        "phase2"
    )


    if (
        cell102_phase2_root.is_dir()
    ):

        cell102_filename_matches = [
            path.resolve()

            for path
            in cell102_phase2_root.rglob(
                f"*{cell101_source_token}*"
            )

            if (
                path.is_file()

                and

                path.resolve()
                !=
                cell101_accepted_corpus_path
            )
        ]


cell102_candidate_files.extend(
    cell102_filename_matches
)


cell102_candidate_files = sorted(
    set(
        cell102_candidate_files
    )
)


# ============================================================
# 8) Read candidate upstream artifacts
# ============================================================

CELL102_TERMS = [
    "شوک‌های درآمدی",
    "هزین‌های",
    "هزینه",
    "شغل و درآمد",
    "شغل",
    "چهارده مفهوم سواد مالی",
]


cell102_artifact_inspections = []


for path in (
    cell102_candidate_files
):

    try:

        text = (
            path.read_text(
                encoding="utf-8",
                errors="replace",
            )
        )

    except Exception as exc:

        cell102_artifact_inspections.append(
            {
                "path":
                    str(
                        path
                    ),

                "read_status":
                    "FAILED",

                "error":
                    repr(
                        exc
                    ),

                "term_hits":
                    {},
            }
        )

        continue


    term_hits = {
        term:
            text.count(
                term
            )

        for term
        in CELL102_TERMS
    }


    cell102_artifact_inspections.append(
        {
            "path":
                str(
                    path
                ),

            "read_status":
                "PASS",

            "term_hits":
                term_hits,

            "contains_canonical_url":
                (
                    cell101_canonical_url
                    in
                    text
                ),

            "contains_source_token":
                (
                    cell101_source_token
                    in
                    text
                ),
        }
    )


# ============================================================
# 9) Keep only artifacts linked to exact Source
# ============================================================

cell102_exact_source_artifacts = [
    item

    for item
    in cell102_artifact_inspections

    if (
        item.get(
            "read_status"
        )
        ==
        "PASS"

        and

        (
            item.get(
                "contains_canonical_url"
            )
            is True

            or

            item.get(
                "contains_source_token"
            )
            is True

            or

            cell101_source_token
            in
            Path(
                item[
                    "path"
                ]
            ).name
        )
    )
]


# ============================================================
# 10) Determine evidence state without correcting anything
# ============================================================

cell102_upstream_has_current_wording = any(
    (
        item[
            "term_hits"
        ].get(
            "شوک‌های درآمدی",
            0,
        )
        >
        0

        and

        item[
            "term_hits"
        ].get(
            "هزین‌های",
            0,
        )
        >
        0
    )

    for item
    in cell102_exact_source_artifacts
)


cell102_upstream_has_job_income = any(
    item[
        "term_hits"
    ].get(
        "شغل و درآمد",
        0,
    )
    >
    0

    for item
    in cell102_exact_source_artifacts
)


cell102_upstream_has_complete_cost_term = any(
    item[
        "term_hits"
    ].get(
        "هزینه",
        0,
    )
    >
    0

    for item
    in cell102_exact_source_artifacts
)


if (
    len(
        cell102_exact_source_artifacts
    )
    ==
    0
):

    cell102_interpretation = (
        "UPSTREAM_INTERNAL_ARTIFACT_NOT_LOCATED"
    )

    cell102_next_action = (
        "KEEP_EXPECTED_ANSWER_BLOCKED_AND_DECIDE_IF_ADDITIONAL_SOURCE_VERIFICATION_IS_ALLOWED"
    )


elif (
    cell102_upstream_has_job_income
    or
    cell102_upstream_has_complete_cost_term
):

    cell102_interpretation = (
        "UPSTREAM_INTERNAL_TEXT_DIFFERS_OR_CONTAINS_ADDITIONAL_WORDING"
    )

    cell102_next_action = (
        "COMPARE_EXACT_UPSTREAM_TEXT_WITH_ACCEPTED_CORPUS_BEFORE_ANY_REPAIR"
    )


elif (
    cell102_upstream_has_current_wording
):

    cell102_interpretation = (
        "AMBIGUOUS_WORDING_IS_PRESENT_IN_UPSTREAM_INTERNAL_ARTIFACT"
    )

    cell102_next_action = (
        "DO_NOT_CORRECT_BY_INFERENCE_KEEP_EXPECTED_ANSWER_BLOCKED"
    )


else:

    cell102_interpretation = (
        "UPSTREAM_ARTIFACT_FOUND_BUT_DISPUTED_WORDING_NOT_RESOLVED"
    )

    cell102_next_action = (
        "KEEP_EXPECTED_ANSWER_BLOCKED_PENDING_FURTHER_VERIFICATION"
    )


# ============================================================
# 11) Verification record
# ============================================================

cell102_upstream_verification_record = {
    "verification_version":
        "same_source_upstream_internal_artifact_inspection_v1",

    "pilot_case_id":
        cell100_human_adjudication_record[
            "pilot_case_id"
        ],

    "source_token":
        cell101_source_token,

    "canonical_url":
        cell101_canonical_url,

    "candidate_file_count":
        len(
            cell102_candidate_files
        ),

    "exact_source_artifact_count":
        len(
            cell102_exact_source_artifacts
        ),

    "upstream_has_current_wording":
        cell102_upstream_has_current_wording,

    "upstream_has_job_income":
        cell102_upstream_has_job_income,

    "upstream_has_complete_cost_term":
        cell102_upstream_has_complete_cost_term,

    "interpretation":
        cell102_interpretation,

    "evidence_corrected":
        False,

    "user_hypothesis_promoted_to_fact":
        False,

    "expected_answer_generated":
        False,

    "next_action":
        cell102_next_action,
}


cell102_upstream_verification_sha256 = (
    cell102_sha(
        cell102_upstream_verification_record
    )
)


# ============================================================
# 12) Freeze protected state AFTER inspection
# ============================================================

cell102_source_record_snapshot_after = (
    cell102_sha(
        cell101_source_record
    )
)


cell102_question_snapshot_after = (
    cell102_sha(
        cell97_question_draft_record
    )
)


cell102_adjudication_snapshot_after = (
    cell102_sha(
        cell100_human_adjudication_record
    )
)


cell102_pilot_runtime_snapshot_after = (
    cell102_sha(
        {
            "question_drafting_pilot_packets":
                question_drafting_pilot_packets,

            "question_drafting_pilot_by_id":
                question_drafting_pilot_by_id,
        }
    )
)


# ============================================================
# 13) Final gates
# ============================================================

cell102_gates = {
    "cell101_is_pass":
        (
            globals().get(
                "cell_101_pass"
            )
            is True
        ),

    "repository_root_is_resolved":
        (
            cell102_repo_root
            is not None
        ),

    "inspection_completed_without_correction":
        (
            cell102_upstream_verification_record[
                "evidence_corrected"
            ]
            is False
        ),

    "human_hypothesis_not_promoted_to_fact":
        (
            cell102_upstream_verification_record[
                "user_hypothesis_promoted_to_fact"
            ]
            is False
        ),

    "expected_answer_not_generated":
        (
            cell102_upstream_verification_record[
                "expected_answer_generated"
            ]
            is False
        ),

    "source_record_remains_byte_stable":
        (
            cell102_source_record_snapshot_before
            ==
            cell102_source_record_snapshot_after
        ),

    "question_record_remains_byte_stable":
        (
            cell102_question_snapshot_before
            ==
            cell102_question_snapshot_after
        ),

    "adjudication_record_remains_byte_stable":
        (
            cell102_adjudication_snapshot_before
            ==
            cell102_adjudication_snapshot_after
        ),

    "pilot_runtime_remains_byte_stable":
        (
            cell102_pilot_runtime_snapshot_before
            ==
            cell102_pilot_runtime_snapshot_after
        ),

    "no_artifact_write":
        True,
}


cell_102_pass = all(
    cell102_gates.values()
)


cell102_status = (
    "UPSTREAM_INTERNAL_SOURCE_INSPECTED_PENDING_DECISION"

    if (
        cell_102_pass
    )

    else
    "UPSTREAM_INTERNAL_SOURCE_INSPECTION_FAILED"
)


# ============================================================
# 14) Diagnostics
# ============================================================

print(
    "=== UPSTREAM INTERNAL SOURCE INSPECTION ==="
)


print(
    "\n=== DISCOVERY ==="
)

print(
    "Repository root:",
    cell102_repo_root
)

print(
    "Lineage/path references found:",
    len(
        cell102_lineage_candidates
    )
)

print(
    "Candidate files:",
    len(
        cell102_candidate_files
    )
)

print(
    "Exact-source artifacts:",
    len(
        cell102_exact_source_artifacts
    )
)


print(
    "\n=== EXACT SOURCE ARTIFACTS ==="
)


if (
    cell102_exact_source_artifacts
):

    for index, item in enumerate(
        cell102_exact_source_artifacts,
        start=1,
    ):

        print(
            f"\n--- Artifact {index} ---"
        )

        print(
            "Path:",
            item[
                "path"
            ]
        )

        print(
            "Term hits:",
            item[
                "term_hits"
            ]
        )


else:

    print(
        "No upstream internal artifact linked to this source was located."
    )


print(
    "\n=== VERIFICATION RESULT ==="
)

print(
    "Current wording present upstream:",
    cell102_upstream_has_current_wording
)

print(
    "'شغل و درآمد' present upstream:",
    cell102_upstream_has_job_income
)

print(
    "Complete 'هزینه' wording present upstream:",
    cell102_upstream_has_complete_cost_term
)

print(
    "Interpretation:",
    cell102_interpretation
)

print(
    "Next action:",
    cell102_next_action
)

print(
    "Verification SHA256:",
    cell102_upstream_verification_sha256
)


print(
    "\n=== FAILURES ==="
)

print(
    "Prerequisite failures:",
    cell102_prerequisite_failures
)


print(
    "\nSTATUS:",
    cell102_status
)


print(
    "\n=== CODE CELL 102 GATES ==="
)


for gate_name, gate_value in (
    cell102_gates.items()
):

    print(
        gate_name,
        ":",
        (
            "PASS"
            if gate_value
            else "FAIL"
        )
    )


print(
    "\nFAILED GATES:",
    [
        gate_name

        for gate_name, gate_value
        in cell102_gates.items()

        if not gate_value
    ]
)


print(
    "\nCODE CELL 102 GATE:",
    (
        "PASS"
        if cell_102_pass
        else "FAIL"
    )
)


assert cell_102_pass, (
    "Upstream internal source inspection failed. "
    "Do not correct Evidence or generate an Expected Answer. "
    "Repair this SAME Code Cell 102 using the printed diagnostics."
)
```

```python
# Code Cell 103

# ============================================================
# Inspect exact Raw HTML contexts around disputed terms.
#
# No correction.
# No evidence mutation.
# No expected-answer generation.
# No artifact write.
# ============================================================


# ============================================================
# 1) Prerequisites
# ============================================================

cell103_prerequisite_failures = []


if (
    globals().get(
        "cell_102_pass"
    )
    is not True
):

    cell103_prerequisite_failures.append(
        "CELL102_NOT_PASS"
    )


if (
    globals().get(
        "cell102_status"
    )
    !=
    "UPSTREAM_INTERNAL_SOURCE_INSPECTED_PENDING_DECISION"
):

    cell103_prerequisite_failures.append(
        {
            "reason":
                "CELL102_STATUS_UNEXPECTED",

            "actual":
                globals().get(
                    "cell102_status"
                ),
        }
    )


CELL103_REQUIRED_RUNTIME_OBJECTS = [
    "cell102_exact_source_artifacts",
    "cell101_source_record",
    "cell97_question_draft_record",
    "cell100_human_adjudication_record",
]


cell103_missing_runtime_objects = [
    name

    for name
    in CELL103_REQUIRED_RUNTIME_OBJECTS

    if (
        name
        not in
        globals()
    )
]


if (
    cell103_missing_runtime_objects
):

    cell103_prerequisite_failures.append(
        {
            "reason":
                "REQUIRED_RUNTIME_OBJECTS_MISSING",

            "objects":
                cell103_missing_runtime_objects,
        }
    )


# ============================================================
# 2) Helper
# ============================================================

def cell103_sha(value):

    return hashlib.sha256(
        stable_json_key(
            value
        ).encode(
            "utf-8"
        )
    ).hexdigest()


def cell103_find_all(text, term):

    positions = []

    start = 0

    while True:

        index = text.find(
            term,
            start,
        )

        if (
            index
            ==
            -1
        ):

            break

        positions.append(
            index
        )

        start = (
            index
            +
            len(
                term
            )
        )

    return positions


def cell103_context(text, index, term, radius=220):

    start = max(
        0,
        index - radius,
    )

    end = min(
        len(
            text
        ),
        index
        +
        len(
            term
        )
        +
        radius,
    )

    return (
        text[
            start:end
        ]
        .replace(
            "\r",
            " ",
        )
        .replace(
            "\n",
            " ",
        )
    )


# ============================================================
# 3) Freeze protected runtime
# ============================================================

cell103_source_snapshot_before = (
    cell103_sha(
        cell101_source_record
    )
)


cell103_question_snapshot_before = (
    cell103_sha(
        cell97_question_draft_record
    )
)


cell103_adjudication_snapshot_before = (
    cell103_sha(
        cell100_human_adjudication_record
    )
)


cell103_pilot_runtime_snapshot_before = (
    cell103_sha(
        {
            "question_drafting_pilot_packets":
                question_drafting_pilot_packets,

            "question_drafting_pilot_by_id":
                question_drafting_pilot_by_id,
        }
    )
)


# ============================================================
# 4) Resolve exact Raw HTML artifact
# ============================================================

cell103_raw_candidates = [
    Path(
        item[
            "path"
        ]
    )

    for item
    in cell102_exact_source_artifacts

    if (
        Path(
            item[
                "path"
            ]
        ).name
        ==
        "article.html"
    )
]


cell103_raw_failures = []


if (
    len(
        cell103_raw_candidates
    )
    !=
    1
):

    cell103_raw_failures.append(
        {
            "reason":
                "RAW_HTML_NOT_UNIQUE",

            "count":
                len(
                    cell103_raw_candidates
                ),

            "candidates":
                [
                    str(
                        path
                    )

                    for path
                    in cell103_raw_candidates
                ],
        }
    )


cell103_raw_path = (
    cell103_raw_candidates[
        0
    ]

    if (
        len(
            cell103_raw_candidates
        )
        ==
        1
    )

    else None
)


cell103_raw_html = (
    cell103_raw_path.read_text(
        encoding="utf-8",
        errors="replace",
    )

    if (
        cell103_raw_path
        is not None
    )

    else ""
)


# ============================================================
# 5) Inspect every relevant occurrence
# ============================================================

CELL103_TERMS = [
    "شوک‌های درآمدی",
    "هزین‌های",
    "هزینه",
    "شغل و درآمد",
    "شغل",
    "درآمد",
    "چهارده مفهوم سواد مالی",
]


cell103_occurrences = {}


for term in (
    CELL103_TERMS
):

    positions = (
        cell103_find_all(
            cell103_raw_html,
            term,
        )
    )

    cell103_occurrences[
        term
    ] = [
        {
            "index":
                index,

            "context":
                cell103_context(
                    cell103_raw_html,
                    index,
                    term,
                ),
        }

        for index
        in positions
    ]


# ============================================================
# 6) Inspect exact financial-literacy sentence in Raw HTML
# ============================================================

CELL103_CURRENT_TARGET_FRAGMENT = (
    "از خرج و پس‌انداز و شوک‌های درآمدی و هزین‌های گرفته تا "
    "اعتبار و بدهی، سرمایه‌گذاری و مدیریت ریسک و تصمیم‌گیری مالی "
    "که به شماره تا چهارده مفهوم سواد مالی می‌رسد"
)


cell103_target_fragment_positions = (
    cell103_find_all(
        cell103_raw_html,
        CELL103_CURRENT_TARGET_FRAGMENT,
    )
)


cell103_target_fragment_contexts = [
    {
        "index":
            index,

        "context":
            cell103_context(
                cell103_raw_html,
                index,
                CELL103_CURRENT_TARGET_FRAGMENT,
                radius=350,
            ),
    }

    for index
    in cell103_target_fragment_positions
]


# ============================================================
# 7) Determine whether extra terms belong to same local context
# ============================================================

cell103_target_start = (
    cell103_target_fragment_positions[
        0
    ]

    if (
        len(
            cell103_target_fragment_positions
        )
        ==
        1
    )

    else None
)


cell103_target_end = (
    cell103_target_start
    +
    len(
        CELL103_CURRENT_TARGET_FRAGMENT
    )

    if (
        cell103_target_start
        is not None
    )

    else None
)


def cell103_term_inside_target(term):

    if (
        cell103_target_start
        is None
    ):

        return None

    return any(
        (
            position
            >=
            cell103_target_start

            and

            position
            <
            cell103_target_end
        )

        for position
        in cell103_find_all(
            cell103_raw_html,
            term,
        )
    )


cell103_same_fragment_membership = {
    term:
        cell103_term_inside_target(
            term
        )

    for term
    in [
        "شوک‌های درآمدی",
        "هزین‌های",
        "هزینه",
        "شغل و درآمد",
        "شغل",
    ]
}


# ============================================================
# 8) Read-only verification record
# ============================================================

cell103_context_verification_record = {
    "verification_version":
        "raw_html_disputed_term_context_inspection_v1",

    "raw_path":
        (
            str(
                cell103_raw_path
            )

            if (
                cell103_raw_path
                is not None
            )

            else None
        ),

    "target_fragment_match_count":
        len(
            cell103_target_fragment_positions
        ),

    "term_occurrence_counts":
        {
            term:
                len(
                    items
                )

            for term, items
            in cell103_occurrences.items()
        },

    "same_target_fragment_membership":
        dict(
            cell103_same_fragment_membership
        ),

    "evidence_modified":
        False,

    "question_modified":
        False,

    "expected_answer_generated":
        False,
}


cell103_context_verification_sha256 = (
    cell103_sha(
        cell103_context_verification_record
    )
)


# ============================================================
# 9) Freeze protected runtime AFTER inspection
# ============================================================

cell103_source_snapshot_after = (
    cell103_sha(
        cell101_source_record
    )
)


cell103_question_snapshot_after = (
    cell103_sha(
        cell97_question_draft_record
    )
)


cell103_adjudication_snapshot_after = (
    cell103_sha(
        cell100_human_adjudication_record
    )
)


cell103_pilot_runtime_snapshot_after = (
    cell103_sha(
        {
            "question_drafting_pilot_packets":
                question_drafting_pilot_packets,

            "question_drafting_pilot_by_id":
                question_drafting_pilot_by_id,
        }
    )
)


# ============================================================
# 10) Final gates
# ============================================================

cell103_gates = {
    "cell102_is_pass":
        (
            globals().get(
                "cell_102_pass"
            )
            is True
        ),

    "raw_html_is_unique":
        (
            len(
                cell103_raw_failures
            )
            ==
            0
        ),

    "target_fragment_found_exactly_once":
        (
            len(
                cell103_target_fragment_positions
            )
            ==
            1
        ),

    "inspection_is_read_only":
        (
            cell103_context_verification_record[
                "evidence_modified"
            ]
            is False

            and

            cell103_context_verification_record[
                "question_modified"
            ]
            is False
        ),

    "no_expected_answer_generated":
        (
            cell103_context_verification_record[
                "expected_answer_generated"
            ]
            is False
        ),

    "source_record_remains_byte_stable":
        (
            cell103_source_snapshot_before
            ==
            cell103_source_snapshot_after
        ),

    "question_record_remains_byte_stable":
        (
            cell103_question_snapshot_before
            ==
            cell103_question_snapshot_after
        ),

    "adjudication_record_remains_byte_stable":
        (
            cell103_adjudication_snapshot_before
            ==
            cell103_adjudication_snapshot_after
        ),

    "pilot_runtime_remains_byte_stable":
        (
            cell103_pilot_runtime_snapshot_before
            ==
            cell103_pilot_runtime_snapshot_after
        ),

    "no_artifact_write":
        True,
}


cell_103_pass = all(
    cell103_gates.values()
)


cell103_status = (
    "RAW_HTML_TERM_CONTEXTS_INSPECTED_PENDING_HUMAN_INTERPRETATION"

    if (
        cell_103_pass
    )

    else
    "RAW_HTML_TERM_CONTEXT_INSPECTION_FAILED"
)


# ============================================================
# 11) Diagnostics
# ============================================================

print(
    "=== RAW HTML — DISPUTED TERM CONTEXT INSPECTION ==="
)


print(
    "\nRaw HTML:",
    cell103_raw_path
)


print(
    "\n=== TARGET FRAGMENT ==="
)

print(
    "Exact fragment matches:",
    len(
        cell103_target_fragment_positions
    )
)


for item in (
    cell103_target_fragment_contexts
):

    print(
        "\nTarget fragment context:"
    )

    print(
        item[
            "context"
        ]
    )


print(
    "\n=== TERM OCCURRENCES ==="
)


for term in (
    CELL103_TERMS
):

    items = (
        cell103_occurrences[
            term
        ]
    )

    print(
        "\nTERM:",
        repr(
            term
        )
    )

    print(
        "Count:",
        len(
            items
        )
    )


    for occurrence_index, item in enumerate(
        items,
        start=1,
    ):

        print(
            f"\nOccurrence {occurrence_index}"
        )

        print(
            "Index:",
            item[
                "index"
            ]
        )

        print(
            "Context:"
        )

        print(
            item[
                "context"
            ]
        )


print(
    "\n=== SAME TARGET FRAGMENT MEMBERSHIP ==="
)


for term, inside in (
    cell103_same_fragment_membership.items()
):

    print(
        repr(
            term
        ),
        ":",
        inside
    )


print(
    "\n=== VERIFICATION BOUNDARY ==="
)

print(
    "Evidence modified:",
    cell103_context_verification_record[
        "evidence_modified"
    ]
)

print(
    "Question modified:",
    cell103_context_verification_record[
        "question_modified"
    ]
)

print(
    "Expected answer generated:",
    cell103_context_verification_record[
        "expected_answer_generated"
    ]
)

print(
    "Verification SHA256:",
    cell103_context_verification_sha256
)


print(
    "\n=== FAILURES ==="
)

print(
    "Prerequisite failures:",
    cell103_prerequisite_failures
)

print(
    "Raw failures:",
    cell103_raw_failures
)


print(
    "\nSTATUS:",
    cell103_status
)


print(
    "\n=== CODE CELL 103 GATES ==="
)


for gate_name, gate_value in (
    cell103_gates.items()
):

    print(
        gate_name,
        ":",
        (
            "PASS"
            if gate_value
            else "FAIL"
        )
    )


print(
    "\nFAILED GATES:",
    [
        gate_name

        for gate_name, gate_value
        in cell103_gates.items()

        if not gate_value
    ]
)


print(
    "\nCODE CELL 103 GATE:",
    (
        "PASS"
        if cell_103_pass
        else "FAIL"
    )
)


assert cell_103_pass, (
    "Raw HTML context inspection failed. "
    "Do not repair Evidence or generate an Expected Answer. "
    "Repair this SAME Code Cell 103 using the printed diagnostics."
)
```

<div style="direction: rtl; white-space: normal; line-height: 1.7;">

<h3>تفسیر نهایی Evidence Verification و تعیین مسیر Repair</h3>

<p>
بررسی Raw HTML نشان داد عبارت‌های
<code>شوک‌های درآمدی</code>
و
<code>هزین‌های</code>
دقیقاً در متن اصلی همان مقاله وجود دارند.
بنابراین این دو عبارت در Pipeline ساخته نشده‌اند.
</p>

<p>
همچنین مشخص شد:
</p>

<ul>
<li><code>هزینه</code> در بخش دیگری از صفحه و در Navigation با عبارت <code>کاهش هزینه‌ها</code> دیده می‌شود و ادامه عبارت ناقص مقاله نیست.</li>
<li><code>شغل</code> مربوط به معرفی مقالۀ دیگری در همان صفحه است.</li>
<li><code>شغل و درآمد</code> اصلاً در فایل وجود ندارد.</li>
<li><code>کاریابی و درآمد</code> در Category/Navigation سایت دیده می‌شود، اما بخشی از متن این مقاله و Target Evidence نیست.</li>
</ul>

<p>
در نتیجه جایگزین‌کردن <code>شوک‌های درآمدی</code> با <code>شغل و درآمد</code> یا <code>کاریابی و درآمد</code> مجاز نیست.
همچنین تکمیل حدسی <code>هزین‌های</code> به یک عبارت خاص ممنوع است.
</p>

<p>
برای یک Golden QA از نوع <code>EVIDENCE_COMPLETENESS</code> این Target Evidence کیفیت کافی برای Expected Answer دقیق ندارد، زیرا سؤال باید فهرست کامل موضوعات صریح را ارزیابی کند اما یکی از اقلام Source ناقص است.
</p>

<p>
بنابراین مسیر پیشنهادی Repair:
<code>RETARGET_PILOT_EVIDENCE</code>
است؛ یعنی Question و Pilot حفظ شوند، اما برای همین Task یک Target Evidence سالم و بدون ابهام انتخاب شود.
</p>

<ul>
<li>هیچ Evidence فعلی Rewrite نمی‌شود.</li>
<li>هیچ Correction حدسی انجام نمی‌شود.</li>
<li>Expected Answer همچنان Blocked می‌ماند.</li>
<li>Full 572 generation همچنان Blocked است.</li>
<li>هیچ File Write یا Artifact Write انجام نمی‌شود.</li>
</ul>

<h4>معیار PASS</h4>

<p>
نتیجه باید صریحاً Source typo/ambiguity را از Pipeline error جدا کند،
فرض <code>شغل و درآمد</code> را Unsupported نگه دارد،
Target فعلی را برای <code>EVIDENCE_COMPLETENESS</code> نامناسب تشخیص دهد
و فقط مسیر <code>RETARGET_PILOT_EVIDENCE</code> را Stage کند؛ بدون Mutation.
</p>

</div>

```python
# Code Cell 104

# ============================================================
# Interpret completed same-source verification.
#
# No evidence correction.
# No question rewrite.
# No expected-answer generation.
# No artifact write.
# ============================================================


# ============================================================
# 1) Prerequisites
# ============================================================

cell104_prerequisite_failures = []


if (
    globals().get(
        "cell_103_pass"
    )
    is not True
):

    cell104_prerequisite_failures.append(
        "CELL103_NOT_PASS"
    )


if (
    globals().get(
        "cell103_status"
    )
    !=
    "RAW_HTML_TERM_CONTEXTS_INSPECTED_PENDING_HUMAN_INTERPRETATION"
):

    cell104_prerequisite_failures.append(
        {
            "reason":
                "CELL103_STATUS_UNEXPECTED",

            "actual":
                globals().get(
                    "cell103_status"
                ),
        }
    )


CELL104_REQUIRED_RUNTIME_OBJECTS = [
    "cell103_context_verification_record",
    "cell103_same_fragment_membership",
    "cell103_occurrences",
    "cell100_human_adjudication_record",
    "cell97_question_draft_record",
    "cell96_next_pilot_standard_workflow_stage",
]


cell104_missing_runtime_objects = [
    name

    for name
    in CELL104_REQUIRED_RUNTIME_OBJECTS

    if (
        name
        not in
        globals()
    )
]


if (
    cell104_missing_runtime_objects
):

    cell104_prerequisite_failures.append(
        {
            "reason":
                "REQUIRED_RUNTIME_OBJECTS_MISSING",

            "objects":
                cell104_missing_runtime_objects,
        }
    )


# ============================================================
# 2) Helper
# ============================================================

def cell104_sha(value):

    return hashlib.sha256(
        stable_json_key(
            value
        ).encode(
            "utf-8"
        )
    ).hexdigest()


# ============================================================
# 3) Freeze protected state BEFORE interpretation
# ============================================================

cell104_question_snapshot_before = (
    cell104_sha(
        cell97_question_draft_record
    )
)


cell104_staging_snapshot_before = (
    cell104_sha(
        cell96_next_pilot_standard_workflow_stage
    )
)


cell104_adjudication_snapshot_before = (
    cell104_sha(
        cell100_human_adjudication_record
    )
)


cell104_pilot_snapshot_before = (
    cell104_sha(
        {
            "question_drafting_pilot_packets":
                question_drafting_pilot_packets,

            "question_drafting_pilot_by_id":
                question_drafting_pilot_by_id,
        }
    )
)


cell104_queue_snapshot_before = (
    cell104_sha(
        cell95_remaining_pilot_queue
    )
)


# ============================================================
# 4) Evidence-backed interpretation
# ============================================================

cell104_interpretation_failures = []


cell104_current_wording_confirmed_in_raw = (
    cell103_same_fragment_membership.get(
        "شوک‌های درآمدی"
    )
    is True

    and

    cell103_same_fragment_membership.get(
        "هزین‌های"
    )
    is True
)


cell104_complete_cost_is_same_fragment = (
    cell103_same_fragment_membership.get(
        "هزینه"
    )
    is True
)


cell104_job_income_is_same_fragment = (
    cell103_same_fragment_membership.get(
        "شغل و درآمد"
    )
    is True
)


cell104_job_is_same_fragment = (
    cell103_same_fragment_membership.get(
        "شغل"
    )
    is True
)


cell104_job_income_exact_occurrences = (
    len(
        cell103_occurrences.get(
            "شغل و درآمد",
            [],
        )
    )
)


if not (
    cell104_current_wording_confirmed_in_raw
):

    cell104_interpretation_failures.append(
        "CURRENT_ARTICLE_WORDING_NOT_CONFIRMED_IN_RAW"
    )


if (
    cell104_complete_cost_is_same_fragment
):

    cell104_interpretation_failures.append(
        "COMPLETE_COST_TERM_UNEXPECTEDLY_INSIDE_TARGET_FRAGMENT"
    )


if (
    cell104_job_income_is_same_fragment
):

    cell104_interpretation_failures.append(
        "JOB_INCOME_UNEXPECTEDLY_INSIDE_TARGET_FRAGMENT"
    )


if (
    cell104_job_is_same_fragment
):

    cell104_interpretation_failures.append(
        "JOB_TERM_UNEXPECTEDLY_INSIDE_TARGET_FRAGMENT"
    )


if (
    cell104_job_income_exact_occurrences
    !=
    0
):

    cell104_interpretation_failures.append(
        "JOB_INCOME_HYPOTHESIS_HAS_EXACT_RAW_SUPPORT_UNEXPECTEDLY"
    )


# ============================================================
# 5) Classify verified source-quality issue
# ============================================================

cell104_source_quality_classification = {
    "current_article_wording":
        "CONFIRMED_IN_RAW_SOURCE",

    "shock_income_phrase":
        "SOURCE_TEXT_CONFIRMED",

    "malformed_cost_phrase":
        "SOURCE_TEXT_MALFORMED_OR_INCOMPLETE_CONFIRMED",

    "complete_cost_term_in_same_fragment":
        False,

    "job_income_hypothesis":
        "NOT_SUPPORTED_BY_ARTICLE_TARGET_TEXT",

    "job_term_in_same_fragment":
        False,

    "pipeline_extraction_error":
        "NOT_SUPPORTED",

    "safe_automatic_correction_available":
        False,
}


# ============================================================
# 6) Evaluate suitability for current Question archetype
# ============================================================

cell104_question_archetype = (
    cell96_next_pilot_standard_workflow_stage[
        "question_archetype"
    ]
)


cell104_target_suitability_failures = []


if (
    cell104_question_archetype
    !=
    "EVIDENCE_COMPLETENESS"
):

    cell104_target_suitability_failures.append(
        {
            "reason":
                "QUESTION_ARCHETYPE_UNEXPECTED",

            "actual":
                cell104_question_archetype,
        }
    )


cell104_current_target_suitable_for_expected_answer = (
    False
)


cell104_current_target_suitability_reason = (
    "The question requires complete enumeration of explicitly "
    "stated financial-literacy topics, but one source item is "
    "malformed/incomplete in the original article text. "
    "A clean Expected Answer would require unsupported correction."
)


# ============================================================
# 7) Stage repair route only
# ============================================================

cell104_repair_route = {
    "repair_route_version":
        "pilot_target_evidence_repair_route_v1",

    "pilot_case_id":
        cell100_human_adjudication_record[
            "pilot_case_id"
        ],

    "question_draft_id":
        cell97_question_draft_record[
            "question_draft_id"
        ],

    "question_text":
        cell97_question_draft_record[
            "question_text"
        ],

    "question_status":
        "RETAIN_CURRENT_DRAFT",

    "current_target_evidence_id":
        cell100_human_adjudication_record[
            "target_evidence_id"
        ],

    "current_target_status":
        "UNSUITABLE_FOR_CLEAN_EVIDENCE_COMPLETENESS_EXPECTED_ANSWER",

    "source_quality_classification":
        dict(
            cell104_source_quality_classification
        ),

    "user_hypothesis":
        "Possible intended concept: شغل و درآمد",

    "user_hypothesis_status":
        "NOT_SUPPORTED_BY_CURRENT_ARTICLE_EVIDENCE",

    "automatic_source_correction":
        "FORBIDDEN",

    "expected_answer_status":
        "BLOCKED_PENDING_RETARGETING",

    "recommended_repair_action":
        "RETARGET_PILOT_EVIDENCE",

    "repair_constraint":
        (
            "Use only existing internal validated evidence; "
            "do not invent or silently correct source wording."
        ),

    "full_572_generation":
        "BLOCKED_PENDING_FULL_PILOT_WORKFLOW_VALIDATION",

    "next_action":
        "DISCOVER_ALTERNATIVE_VALIDATED_TARGET_EVIDENCE_FOR_SAME_PILOT_TASK",
}


cell104_repair_route_sha256 = (
    cell104_sha(
        cell104_repair_route
    )
)


# ============================================================
# 8) Validate repair route
# ============================================================

cell104_route_failures = []


if (
    cell104_repair_route[
        "recommended_repair_action"
    ]
    !=
    "RETARGET_PILOT_EVIDENCE"
):

    cell104_route_failures.append(
        "REPAIR_ACTION_UNEXPECTED"
    )


if (
    cell104_repair_route[
        "automatic_source_correction"
    ]
    !=
    "FORBIDDEN"
):

    cell104_route_failures.append(
        "AUTOMATIC_SOURCE_CORRECTION_NOT_FORBIDDEN"
    )


if (
    cell104_repair_route[
        "expected_answer_status"
    ]
    !=
    "BLOCKED_PENDING_RETARGETING"
):

    cell104_route_failures.append(
        "EXPECTED_ANSWER_NOT_BLOCKED_PENDING_RETARGETING"
    )


if (
    cell104_repair_route[
        "user_hypothesis_status"
    ]
    !=
    "NOT_SUPPORTED_BY_CURRENT_ARTICLE_EVIDENCE"
):

    cell104_route_failures.append(
        "USER_HYPOTHESIS_STATUS_UNEXPECTED"
    )


# ============================================================
# 9) Freeze protected state AFTER interpretation
# ============================================================

cell104_question_snapshot_after = (
    cell104_sha(
        cell97_question_draft_record
    )
)


cell104_staging_snapshot_after = (
    cell104_sha(
        cell96_next_pilot_standard_workflow_stage
    )
)


cell104_adjudication_snapshot_after = (
    cell104_sha(
        cell100_human_adjudication_record
    )
)


cell104_pilot_snapshot_after = (
    cell104_sha(
        {
            "question_drafting_pilot_packets":
                question_drafting_pilot_packets,

            "question_drafting_pilot_by_id":
                question_drafting_pilot_by_id,
        }
    )
)


cell104_queue_snapshot_after = (
    cell104_sha(
        cell95_remaining_pilot_queue
    )
)


# ============================================================
# 10) Final gates
# ============================================================

cell104_gates = {
    "cell103_is_pass":
        (
            globals().get(
                "cell_103_pass"
            )
            is True
        ),

    "raw_source_confirms_current_article_wording":
        (
            cell104_current_wording_confirmed_in_raw
        ),

    "complete_cost_term_is_not_same_fragment":
        (
            cell104_complete_cost_is_same_fragment
            is False
        ),

    "job_income_is_not_same_fragment":
        (
            cell104_job_income_is_same_fragment
            is False
        ),

    "job_income_hypothesis_is_not_supported":
        (
            cell104_job_income_exact_occurrences
            ==
            0
        ),

    "interpretation_has_no_conflicts":
        (
            len(
                cell104_interpretation_failures
            )
            ==
            0
        ),

    "current_target_is_not_accepted_for_clean_expected_answer":
        (
            cell104_current_target_suitable_for_expected_answer
            is False
        ),

    "repair_route_is_valid":
        (
            len(
                cell104_route_failures
            )
            ==
            0
        ),

    "question_is_retained":
        (
            cell104_repair_route[
                "question_status"
            ]
            ==
            "RETAIN_CURRENT_DRAFT"
        ),

    "automatic_correction_is_forbidden":
        (
            cell104_repair_route[
                "automatic_source_correction"
            ]
            ==
            "FORBIDDEN"
        ),

    "expected_answer_remains_blocked":
        (
            cell104_repair_route[
                "expected_answer_status"
            ]
            ==
            "BLOCKED_PENDING_RETARGETING"
        ),

    "full_572_generation_remains_blocked":
        (
            cell104_repair_route[
                "full_572_generation"
            ]
            ==
            "BLOCKED_PENDING_FULL_PILOT_WORKFLOW_VALIDATION"
        ),

    "question_record_remains_byte_stable":
        (
            cell104_question_snapshot_before
            ==
            cell104_question_snapshot_after
        ),

    "staging_packet_remains_byte_stable":
        (
            cell104_staging_snapshot_before
            ==
            cell104_staging_snapshot_after
        ),

    "adjudication_record_remains_byte_stable":
        (
            cell104_adjudication_snapshot_before
            ==
            cell104_adjudication_snapshot_after
        ),

    "pilot_runtime_remains_byte_stable":
        (
            cell104_pilot_snapshot_before
            ==
            cell104_pilot_snapshot_after
        ),

    "remaining_queue_remains_byte_stable":
        (
            cell104_queue_snapshot_before
            ==
            cell104_queue_snapshot_after
        ),

    "no_expected_answer_generated":
        True,

    "no_artifact_write":
        True,
}


cell_104_pass = all(
    cell104_gates.values()
)


cell104_status = (
    "CURRENT_TARGET_REJECTED_FOR_EXPECTED_ANSWER_RETARGETING_REQUIRED"

    if (
        cell_104_pass
    )

    else
    "SOURCE_VERIFICATION_INTERPRETATION_FAILED"
)


# ============================================================
# 11) Diagnostics
# ============================================================

print(
    "=== SOURCE VERIFICATION — INTERPRETATION ==="
)


print(
    "\n=== VERIFIED FACTS ==="
)

print(
    "Current article wording confirmed in Raw:",
    cell104_current_wording_confirmed_in_raw
)

print(
    "'شوک‌های درآمدی' status:",
    cell104_source_quality_classification[
        "shock_income_phrase"
    ]
)

print(
    "'هزین‌های' status:",
    cell104_source_quality_classification[
        "malformed_cost_phrase"
    ]
)

print(
    "Complete 'هزینه' in same target fragment:",
    cell104_complete_cost_is_same_fragment
)

print(
    "'شغل و درآمد' exact Raw occurrences:",
    cell104_job_income_exact_occurrences
)

print(
    "'شغل و درآمد' in same target fragment:",
    cell104_job_income_is_same_fragment
)

print(
    "'شغل' in same target fragment:",
    cell104_job_is_same_fragment
)


print(
    "\n=== QUALITY INTERPRETATION ==="
)

print(
    "Pipeline extraction error:",
    cell104_source_quality_classification[
        "pipeline_extraction_error"
    ]
)

print(
    "Safe automatic correction available:",
    cell104_source_quality_classification[
        "safe_automatic_correction_available"
    ]
)

print(
    "Current target suitable for clean Expected Answer:",
    cell104_current_target_suitable_for_expected_answer
)

print(
    "Reason:",
    cell104_current_target_suitability_reason
)


print(
    "\n=== REPAIR ROUTE ==="
)

print(
    "Question status:",
    cell104_repair_route[
        "question_status"
    ]
)

print(
    "Current target status:",
    cell104_repair_route[
        "current_target_status"
    ]
)

print(
    "User hypothesis status:",
    cell104_repair_route[
        "user_hypothesis_status"
    ]
)

print(
    "Recommended action:",
    cell104_repair_route[
        "recommended_repair_action"
    ]
)

print(
    "Expected answer:",
    cell104_repair_route[
        "expected_answer_status"
    ]
)

print(
    "Next action:",
    cell104_repair_route[
        "next_action"
    ]
)

print(
    "Repair route SHA256:",
    cell104_repair_route_sha256
)


print(
    "\n=== FAILURES ==="
)

print(
    "Prerequisite failures:",
    cell104_prerequisite_failures
)

print(
    "Interpretation failures:",
    cell104_interpretation_failures
)

print(
    "Target suitability failures:",
    cell104_target_suitability_failures
)

print(
    "Route failures:",
    cell104_route_failures
)


print(
    "\nSTATUS:",
    cell104_status
)


print(
    "\n=== CODE CELL 104 GATES ==="
)


for gate_name, gate_value in (
    cell104_gates.items()
):

    print(
        gate_name,
        ":",
        (
            "PASS"
            if gate_value
            else "FAIL"
        )
    )


print(
    "\nFAILED GATES:",
    [
        gate_name

        for gate_name, gate_value
        in cell104_gates.items()

        if not gate_value
    ]
)


print(
    "\nCODE CELL 104 GATE:",
    (
        "PASS"
        if cell_104_pass
        else "FAIL"
    )
)


assert cell_104_pass, (
    "Source-verification interpretation failed. "
    "Do not retarget or generate an Expected Answer. "
    "Repair this SAME Code Cell 104 using the printed diagnostics."
)
```

```python
# Code Cell 105

# ============================================================
# Discover alternative validated Target Evidence candidates
# for the same Pilot task.
#
# Read-only discovery only.
# No retargeting.
# No question modification.
# No expected-answer generation.
# No artifact write.
# ============================================================


# ============================================================
# 1) Prerequisites
# ============================================================

cell105_prerequisite_failures = []


if (
    globals().get(
        "cell_104_pass"
    )
    is not True
):

    cell105_prerequisite_failures.append(
        "CELL104_NOT_PASS"
    )


if (
    globals().get(
        "cell104_status"
    )
    !=
    "CURRENT_TARGET_REJECTED_FOR_EXPECTED_ANSWER_RETARGETING_REQUIRED"
):

    cell105_prerequisite_failures.append(
        {
            "reason":
                "CELL104_STATUS_UNEXPECTED",

            "actual":
                globals().get(
                    "cell104_status"
                ),
        }
    )


CELL105_REQUIRED_RUNTIME_OBJECTS = [
    "cell104_repair_route",
    "cell96_next_pilot_standard_workflow_stage",
    "cell96_selected_task",
    "task_draft_plan_rows",
    "target_evidence_records",
    "question_drafting_pilot_packets",
    "question_drafting_pilot_by_id",
    "cell97_question_draft_record",
]


cell105_missing_runtime_objects = [
    name

    for name
    in CELL105_REQUIRED_RUNTIME_OBJECTS

    if (
        name
        not in
        globals()
    )
]


if (
    cell105_missing_runtime_objects
):

    cell105_prerequisite_failures.append(
        {
            "reason":
                "REQUIRED_RUNTIME_OBJECTS_MISSING",

            "objects":
                cell105_missing_runtime_objects,
        }
    )


# ============================================================
# 2) Helpers
# ============================================================

def cell105_sha(value):

    return hashlib.sha256(
        stable_json_key(
            value
        ).encode(
            "utf-8"
        )
    ).hexdigest()


def cell105_walk(value, path=()):

    if isinstance(
        value,
        dict,
    ):

        for key, child in value.items():

            yield from cell105_walk(
                child,
                path + (str(key),),
            )

    elif isinstance(
        value,
        (list, tuple),
    ):

        for index, child in enumerate(
            value
        ):

            yield from cell105_walk(
                child,
                path + (str(index),),
            )

    else:

        yield path, value


def cell105_contains_scalar(value, expected):

    return any(
        scalar
        ==
        expected

        for _, scalar
        in cell105_walk(
            value
        )
    )


def cell105_extract_nonempty_strings(value):

    return [
        {
            "path":
                ".".join(
                    path
                ),

            "value":
                scalar,
        }

        for path, scalar
        in cell105_walk(
            value
        )

        if (
            isinstance(
                scalar,
                str,
            )

            and

            scalar.strip()
        )
    ]


# ============================================================
# 3) Freeze protected state BEFORE discovery
# ============================================================

cell105_task_runtime_snapshot_before = (
    cell105_sha(
        task_draft_plan_rows
    )
)


cell105_target_runtime_snapshot_before = (
    cell105_sha(
        target_evidence_records
    )
)


cell105_pilot_runtime_snapshot_before = (
    cell105_sha(
        {
            "question_drafting_pilot_packets":
                question_drafting_pilot_packets,

            "question_drafting_pilot_by_id":
                question_drafting_pilot_by_id,
        }
    )
)


cell105_question_snapshot_before = (
    cell105_sha(
        cell97_question_draft_record
    )
)


cell105_stage_snapshot_before = (
    cell105_sha(
        cell96_next_pilot_standard_workflow_stage
    )
)


cell105_repair_route_snapshot_before = (
    cell105_sha(
        cell104_repair_route
    )
)


# ============================================================
# 4) Resolve current Pilot / Task identity
# ============================================================

cell105_current_pilot_id = (
    cell96_next_pilot_standard_workflow_stage[
        "pilot_case_id"
    ]
)


cell105_current_golden_case_id = (
    cell96_next_pilot_standard_workflow_stage[
        "golden_case_id"
    ]
)


cell105_current_task_draft_id = (
    cell96_next_pilot_standard_workflow_stage[
        "task_draft_id"
    ]
)


cell105_current_target_evidence_id = (
    cell96_next_pilot_standard_workflow_stage[
        "target_evidence_id"
    ]
)


cell105_current_task_family = (
    cell96_next_pilot_standard_workflow_stage[
        "task_family"
    ]
)


cell105_current_question_archetype = (
    cell96_next_pilot_standard_workflow_stage[
        "question_archetype"
    ]
)


cell105_current_source_type = (
    cell96_next_pilot_standard_workflow_stage[
        "source_type"
    ]
)


cell105_current_stratum = (
    cell96_next_pilot_standard_workflow_stage[
        "stratum"
    ]
)


# ============================================================
# 5) Discover related Task Draft rows
# ============================================================

cell105_related_task_rows = []


for row in (
    task_draft_plan_rows
):

    if not isinstance(
        row,
        dict,
    ):

        continue


    same_golden_case = (
        row.get(
            "golden_case_id"
        )
        ==
        cell105_current_golden_case_id
    )


    same_task_draft = (
        row.get(
            "task_draft_id"
        )
        ==
        cell105_current_task_draft_id
    )


    contains_current_case = (
        cell105_contains_scalar(
            row,
            cell105_current_golden_case_id,
        )
    )


    contains_current_task = (
        cell105_contains_scalar(
            row,
            cell105_current_task_draft_id,
        )
    )


    if (
        same_golden_case
        or
        same_task_draft
        or
        contains_current_case
        or
        contains_current_task
    ):

        cell105_related_task_rows.append(
            row
        )


# ============================================================
# 6) Discover related Target Evidence records
# ============================================================

cell105_related_target_records = []


for record in (
    target_evidence_records
):

    if not isinstance(
        record,
        dict,
    ):

        continue


    if (
        record.get(
            "target_evidence_id"
        )
        ==
        cell105_current_target_evidence_id
    ):

        continue


    linked_to_case = (
        record.get(
            "golden_case_id"
        )
        ==
        cell105_current_golden_case_id

        or

        cell105_contains_scalar(
            record,
            cell105_current_golden_case_id,
        )
    )


    linked_to_task = (
        record.get(
            "task_draft_id"
        )
        ==
        cell105_current_task_draft_id

        or

        cell105_contains_scalar(
            record,
            cell105_current_task_draft_id,
        )
    )


    if (
        linked_to_case
        or
        linked_to_task
    ):

        cell105_related_target_records.append(
            record
        )


# ============================================================
# 7) Discover Target IDs referenced by related Task rows
# ============================================================

cell105_related_target_ids_from_tasks = []


for row in (
    cell105_related_task_rows
):

    for path, scalar in cell105_walk(
        row
    ):

        if not (
            isinstance(
                scalar,
                str,
            )

            and

            scalar.startswith(
                "target-evidence:"
            )
        ):

            continue


        if (
            scalar
            ==
            cell105_current_target_evidence_id
        ):

            continue


        if (
            scalar
            not in
            cell105_related_target_ids_from_tasks
        ):

            cell105_related_target_ids_from_tasks.append(
                scalar
            )


# ============================================================
# 8) Add exact Target records referenced by related Tasks
# ============================================================

cell105_existing_related_ids = {
    record.get(
        "target_evidence_id"
    )

    for record
    in cell105_related_target_records
}


for record in (
    target_evidence_records
):

    if not isinstance(
        record,
        dict,
    ):

        continue


    record_id = (
        record.get(
            "target_evidence_id"
        )
    )


    if (
        record_id
        in
        cell105_related_target_ids_from_tasks

        and

        record_id
        not in
        cell105_existing_related_ids
    ):

        cell105_related_target_records.append(
            record
        )

        cell105_existing_related_ids.add(
            record_id
        )


# ============================================================
# 9) Inspect candidate validation / evidence availability
# ============================================================

CELL105_VALIDATION_MARKERS = [
    "SELECTED_AND_VALIDATED",
    "VALIDATED",
    "PASS",
    "ACCEPTED",
    "READY_FOR_QUESTION_DRAFT",
]


cell105_candidate_summaries = []


for record in (
    cell105_related_target_records
):

    record_strings = (
        cell105_extract_nonempty_strings(
            record
        )
    )


    record_values = [
        item[
            "value"
        ]

        for item
        in record_strings
    ]


    validation_hits = sorted(
        {
            marker

            for marker
            in CELL105_VALIDATION_MARKERS

            if (
                marker
                in
                record_values
            )
        }
    )


    text_candidates = [
        item

        for item
        in record_strings

        if (
            "text"
            in
            item[
                "path"
            ].lower()

            and

            len(
                item[
                    "value"
                ].strip()
            )
            >=
            20
        )
    ]


    locator_candidates = [
        item

        for item
        in record_strings

        if any(
            token
            in
            item[
                "path"
            ].lower()

            for token
            in [
                "locator",
                "url",
                "page",
                "heading",
                "paragraph",
                "timestamp",
            ]
        )
    ]


    attribution_candidates = [
        item

        for item
        in record_strings

        if any(
            token
            in
            item[
                "path"
            ].lower()

            for token
            in [
                "author",
                "attribution",
                "authority",
                "speaker",
            ]
        )
    ]


    candidate_text_blob = "\n".join(
        item[
            "value"
        ]

        for item
        in text_candidates
    )


    has_current_malformed_phrase = (
        "هزین‌های"
        in
        candidate_text_blob
    )


    has_current_shock_phrase = (
        "شوک‌های درآمدی"
        in
        candidate_text_blob
    )


    has_job_income_phrase = (
        "شغل و درآمد"
        in
        candidate_text_blob

        or

        "کاریابی و درآمد"
        in
        candidate_text_blob
    )


    cell105_candidate_summaries.append(
        {
            "target_evidence_id":
                record.get(
                    "target_evidence_id"
                ),

            "task_draft_id":
                record.get(
                    "task_draft_id"
                ),

            "golden_case_id":
                record.get(
                    "golden_case_id"
                ),

            "validation_hits":
                validation_hits,

            "has_text":
                (
                    len(
                        text_candidates
                    )
                    >
                    0
                ),

            "text_candidate_count":
                len(
                    text_candidates
                ),

            "locator_candidate_count":
                len(
                    locator_candidates
                ),

            "attribution_candidate_count":
                len(
                    attribution_candidates
                ),

            "contains_malformed_cost_phrase":
                has_current_malformed_phrase,

            "contains_shock_income_phrase":
                has_current_shock_phrase,

            "contains_job_income_phrase":
                has_job_income_phrase,

            "record":
                record,
        }
    )


# ============================================================
# 10) Conservative eligibility classification
# ============================================================

cell105_eligible_candidates = []


for summary in (
    cell105_candidate_summaries
):

    eligible = (
        summary[
            "target_evidence_id"
        ]
        is not None

        and

        summary[
            "target_evidence_id"
        ]
        !=
        cell105_current_target_evidence_id

        and

        summary[
            "has_text"
        ]

        and

        len(
            summary[
                "validation_hits"
            ]
        )
        >
        0

        and

        summary[
            "locator_candidate_count"
        ]
        >
        0

        and

        summary[
            "attribution_candidate_count"
        ]
        >
        0
    )


    if (
        eligible
    ):

        cell105_eligible_candidates.append(
            summary
        )


# ============================================================
# 11) Discovery result only
# ============================================================

cell105_discovery_result = {
    "discovery_version":
        "alternative_validated_target_evidence_discovery_v1",

    "pilot_case_id":
        cell105_current_pilot_id,

    "golden_case_id":
        cell105_current_golden_case_id,

    "task_draft_id":
        cell105_current_task_draft_id,

    "current_target_evidence_id":
        cell105_current_target_evidence_id,

    "task_family":
        cell105_current_task_family,

    "question_archetype":
        cell105_current_question_archetype,

    "source_type":
        cell105_current_source_type,

    "stratum":
        cell105_current_stratum,

    "related_task_row_count":
        len(
            cell105_related_task_rows
        ),

    "related_target_record_count":
        len(
            cell105_related_target_records
        ),

    "eligible_alternative_candidate_count":
        len(
            cell105_eligible_candidates
        ),

    "candidate_target_evidence_ids":
        [
            item[
                "target_evidence_id"
            ]

            for item
            in cell105_eligible_candidates
        ],

    "candidate_selected":
        False,

    "current_target_modified":
        False,

    "question_modified":
        False,

    "expected_answer_generated":
        False,

    "next_action":
        (
            "INSPECT_ELIGIBLE_ALTERNATIVE_TARGET_CANDIDATES"

            if (
                len(
                    cell105_eligible_candidates
                )
                >
                0
            )

            else
            "NO_EXISTING_VALIDATED_ALTERNATIVE_FOUND_KEEP_PILOT_BLOCKED"
        ),
}


cell105_discovery_result_sha256 = (
    cell105_sha(
        cell105_discovery_result
    )
)


# ============================================================
# 12) Freeze protected state AFTER discovery
# ============================================================

cell105_task_runtime_snapshot_after = (
    cell105_sha(
        task_draft_plan_rows
    )
)


cell105_target_runtime_snapshot_after = (
    cell105_sha(
        target_evidence_records
    )
)


cell105_pilot_runtime_snapshot_after = (
    cell105_sha(
        {
            "question_drafting_pilot_packets":
                question_drafting_pilot_packets,

            "question_drafting_pilot_by_id":
                question_drafting_pilot_by_id,
        }
    )
)


cell105_question_snapshot_after = (
    cell105_sha(
        cell97_question_draft_record
    )
)


cell105_stage_snapshot_after = (
    cell105_sha(
        cell96_next_pilot_standard_workflow_stage
    )
)


cell105_repair_route_snapshot_after = (
    cell105_sha(
        cell104_repair_route
    )
)


# ============================================================
# 13) Final gates
# ============================================================

cell105_gates = {
    "cell104_is_pass":
        (
            globals().get(
                "cell_104_pass"
            )
            is True
        ),

    "repair_route_requests_retargeting":
        (
            cell104_repair_route.get(
                "recommended_repair_action"
            )
            ==
            "RETARGET_PILOT_EVIDENCE"
        ),

    "discovery_completed":
        (
            cell105_discovery_result
            is not None
        ),

    "no_candidate_was_selected":
        (
            cell105_discovery_result[
                "candidate_selected"
            ]
            is False
        ),

    "current_target_not_modified":
        (
            cell105_discovery_result[
                "current_target_modified"
            ]
            is False
        ),

    "question_not_modified":
        (
            cell105_discovery_result[
                "question_modified"
            ]
            is False
        ),

    "expected_answer_not_generated":
        (
            cell105_discovery_result[
                "expected_answer_generated"
            ]
            is False
        ),

    "task_runtime_remains_byte_stable":
        (
            cell105_task_runtime_snapshot_before
            ==
            cell105_task_runtime_snapshot_after
        ),

    "target_runtime_remains_byte_stable":
        (
            cell105_target_runtime_snapshot_before
            ==
            cell105_target_runtime_snapshot_after
        ),

    "pilot_runtime_remains_byte_stable":
        (
            cell105_pilot_runtime_snapshot_before
            ==
            cell105_pilot_runtime_snapshot_after
        ),

    "question_record_remains_byte_stable":
        (
            cell105_question_snapshot_before
            ==
            cell105_question_snapshot_after
        ),

    "staging_packet_remains_byte_stable":
        (
            cell105_stage_snapshot_before
            ==
            cell105_stage_snapshot_after
        ),

    "repair_route_remains_byte_stable":
        (
            cell105_repair_route_snapshot_before
            ==
            cell105_repair_route_snapshot_after
        ),

    "no_artifact_write":
        True,
}


cell_105_pass = all(
    cell105_gates.values()
)


cell105_status = (
    "ALTERNATIVE_TARGET_EVIDENCE_DISCOVERY_COMPLETE"

    if (
        cell_105_pass
    )

    else
    "ALTERNATIVE_TARGET_EVIDENCE_DISCOVERY_FAILED"
)


# ============================================================
# 14) Diagnostics
# ============================================================

print(
    "=== ALTERNATIVE TARGET EVIDENCE DISCOVERY ==="
)


print(
    "\n=== CURRENT PILOT ==="
)

print(
    "Pilot ID:",
    cell105_current_pilot_id
)

print(
    "Golden Case ID:",
    cell105_current_golden_case_id
)

print(
    "Task Draft ID:",
    cell105_current_task_draft_id
)

print(
    "Current Target Evidence ID:",
    cell105_current_target_evidence_id
)

print(
    "Task family:",
    cell105_current_task_family
)

print(
    "Question archetype:",
    cell105_current_question_archetype
)


print(
    "\n=== DISCOVERY COUNTS ==="
)

print(
    "Related Task rows:",
    len(
        cell105_related_task_rows
    )
)

print(
    "Related Target records:",
    len(
        cell105_related_target_records
    )
)

print(
    "Eligible alternative candidates:",
    len(
        cell105_eligible_candidates
    )
)


print(
    "\n=== CANDIDATES ==="
)


if (
    cell105_candidate_summaries
):

    for index, item in enumerate(
        cell105_candidate_summaries,
        start=1,
    ):

        print(
            f"\n--- Candidate {index} ---"
        )

        print(
            "Target Evidence ID:",
            item[
                "target_evidence_id"
            ]
        )

        print(
            "Task Draft ID:",
            item[
                "task_draft_id"
            ]
        )

        print(
            "Golden Case ID:",
            item[
                "golden_case_id"
            ]
        )

        print(
            "Validation hits:",
            item[
                "validation_hits"
            ]
        )

        print(
            "Has text:",
            item[
                "has_text"
            ]
        )

        print(
            "Text candidates:",
            item[
                "text_candidate_count"
            ]
        )

        print(
            "Locator candidates:",
            item[
                "locator_candidate_count"
            ]
        )

        print(
            "Attribution candidates:",
            item[
                "attribution_candidate_count"
            ]
        )

        print(
            "Contains malformed 'هزین‌های':",
            item[
                "contains_malformed_cost_phrase"
            ]
        )

        print(
            "Contains 'شوک‌های درآمدی':",
            item[
                "contains_shock_income_phrase"
            ]
        )

        print(
            "Contains job/income wording:",
            item[
                "contains_job_income_phrase"
            ]
        )


else:

    print(
        "No alternative Target Evidence records were discovered."
    )


print(
    "\n=== ELIGIBLE CANDIDATE IDS ==="
)

print(
    cell105_discovery_result[
        "candidate_target_evidence_ids"
    ]
)


print(
    "\nNext action:",
    cell105_discovery_result[
        "next_action"
    ]
)

print(
    "Discovery SHA256:",
    cell105_discovery_result_sha256
)


print(
    "\n=== FAILURES ==="
)

print(
    "Prerequisite failures:",
    cell105_prerequisite_failures
)


print(
    "\nSTATUS:",
    cell105_status
)


print(
    "\n=== CODE CELL 105 GATES ==="
)


for gate_name, gate_value in (
    cell105_gates.items()
):

    print(
        gate_name,
        ":",
        (
            "PASS"
            if gate_value
            else "FAIL"
        )
    )


print(
    "\nFAILED GATES:",
    [
        gate_name

        for gate_name, gate_value
        in cell105_gates.items()

        if not gate_value
    ]
)


print(
    "\nCODE CELL 105 GATE:",
    (
        "PASS"
        if cell_105_pass
        else "FAIL"
    )
)


assert cell_105_pass, (
    "Alternative Target Evidence discovery failed. "
    "Do not retarget or generate an Expected Answer. "
    "Repair this SAME Code Cell 105 using the printed diagnostics."
)
```

<h4>معیار PASS</h4>

<p>
Workflow Contract و Queue بدون تغییر Inspect شوند و تمام Transition cueهای واقعی چاپ شوند.
اگر Contract مسیر مشخصی نداشته باشد، نتیجه باید همان ابهام را گزارش کند و هیچ Transition حدسی ساخته نشود.

```python
# Code Cell 106

# ============================================================
# Read-only inspection of the frozen Pilot QA workflow contract
# after unsuccessful Target Evidence retarget discovery.
#
# No queue mutation.
# No pilot mutation.
# No expected-answer generation.
# No artifact write.
# ============================================================


# ============================================================
# 1) Prerequisites
# ============================================================

cell106_prerequisite_failures = []


if (
    globals().get(
        "cell_105_pass"
    )
    is not True
):

    cell106_prerequisite_failures.append(
        "CELL105_NOT_PASS"
    )


if (
    globals().get(
        "cell105_status"
    )
    !=
    "ALTERNATIVE_TARGET_EVIDENCE_DISCOVERY_COMPLETE"
):

    cell106_prerequisite_failures.append(
        {
            "reason":
                "CELL105_STATUS_UNEXPECTED",

            "actual":
                globals().get(
                    "cell105_status"
                ),
        }
    )


if (
    globals().get(
        "cell105_discovery_result",
        {}
    ).get(
        "next_action"
    )
    !=
    "NO_EXISTING_VALIDATED_ALTERNATIVE_FOUND_KEEP_PILOT_BLOCKED"
):

    cell106_prerequisite_failures.append(
        {
            "reason":
                "CELL105_NEXT_ACTION_UNEXPECTED",

            "actual":
                globals().get(
                    "cell105_discovery_result",
                    {}
                ).get(
                    "next_action"
                ),
        }
    )


CELL106_REQUIRED_RUNTIME_OBJECTS = [
    "PILOT_QA_WORKFLOW_CONTRACT_V1",
    "cell95_remaining_pilot_queue",
    "cell104_repair_route",
    "cell105_discovery_result",
    "cell100_human_adjudication_record",
    "question_drafting_pilot_packets",
    "question_drafting_pilot_by_id",
]


cell106_missing_runtime_objects = [
    name

    for name
    in CELL106_REQUIRED_RUNTIME_OBJECTS

    if (
        name
        not in
        globals()
    )
]


if (
    cell106_missing_runtime_objects
):

    cell106_prerequisite_failures.append(
        {
            "reason":
                "REQUIRED_RUNTIME_OBJECTS_MISSING",

            "objects":
                cell106_missing_runtime_objects,
        }
    )


# ============================================================
# 2) Helpers
# ============================================================

def cell106_sha(value):

    return hashlib.sha256(
        stable_json_key(
            value
        ).encode(
            "utf-8"
        )
    ).hexdigest()


def cell106_walk(value, path=()):

    if isinstance(
        value,
        dict,
    ):

        for key, child in value.items():

            yield from cell106_walk(
                child,
                path + (str(key),),
            )

    elif isinstance(
        value,
        (list, tuple),
    ):

        for index, child in enumerate(
            value
        ):

            yield from cell106_walk(
                child,
                path + (str(index),),
            )

    else:

        yield path, value


# ============================================================
# 3) Freeze protected runtime BEFORE inspection
# ============================================================

cell106_contract_snapshot_before = (
    cell106_sha(
        PILOT_QA_WORKFLOW_CONTRACT_V1
    )
)


cell106_queue_snapshot_before = (
    cell106_sha(
        cell95_remaining_pilot_queue
    )
)


cell106_pilot_runtime_snapshot_before = (
    cell106_sha(
        {
            "question_drafting_pilot_packets":
                question_drafting_pilot_packets,

            "question_drafting_pilot_by_id":
                question_drafting_pilot_by_id,
        }
    )
)


cell106_adjudication_snapshot_before = (
    cell106_sha(
        cell100_human_adjudication_record
    )
)


cell106_repair_snapshot_before = (
    cell106_sha(
        cell104_repair_route
    )
)


cell106_discovery_snapshot_before = (
    cell106_sha(
        cell105_discovery_result
    )
)


# ============================================================
# 4) Inspect workflow contract structure
# ============================================================

cell106_contract_top_level_keys = (
    sorted(
        PILOT_QA_WORKFLOW_CONTRACT_V1.keys()
    )

    if isinstance(
        PILOT_QA_WORKFLOW_CONTRACT_V1,
        dict,
    )

    else []
)


CELL106_TRANSITION_TERMS = [
    "REJECT",
    "REJECTED",
    "REVISION_REQUIRED",
    "BLOCKED",
    "SKIP",
    "SKIPPED",
    "NEXT",
    "ADVANCE",
    "QUEUE",
    "RETRY",
    "REPAIR",
    "RETARGET",
    "ABANDON",
    "UNUSABLE",
    "INVALID",
    "HUMAN",
    "ACCEPT",
]


cell106_contract_transition_hits = []


for path, scalar in cell106_walk(
    PILOT_QA_WORKFLOW_CONTRACT_V1
):

    path_text = (
        ".".join(
            path
        )
    )


    scalar_text = (
        str(
            scalar
        )
    )


    matched_terms = [
        term

        for term
        in CELL106_TRANSITION_TERMS

        if (
            term.lower()
            in
            path_text.lower()

            or

            term.lower()
            in
            scalar_text.lower()
        )
    ]


    if (
        matched_terms
    ):

        cell106_contract_transition_hits.append(
            {
                "path":
                    path_text,

                "value":
                    scalar,

                "matched_terms":
                    matched_terms,
            }
        )


# ============================================================
# 5) Inspect lifecycle stage definitions
# ============================================================

cell106_lifecycle_candidates = []


for path, scalar in cell106_walk(
    PILOT_QA_WORKFLOW_CONTRACT_V1
):

    path_text = (
        ".".join(
            path
        ).lower()
    )


    if (
        "stage"
        in
        path_text

        or

        "lifecycle"
        in
        path_text
    ):

        cell106_lifecycle_candidates.append(
            {
                "path":
                    ".".join(
                        path
                    ),

                "value":
                    scalar,
            }
        )


# ============================================================
# 6) Inspect current remaining queue
# ============================================================

cell106_queue_count = (
    len(
        cell95_remaining_pilot_queue
    )

    if isinstance(
        cell95_remaining_pilot_queue,
        list,
    )

    else None
)


cell106_current_pilot_id = (
    cell100_human_adjudication_record[
        "pilot_case_id"
    ]
)


cell106_queue_entries_with_current_pilot = []


if isinstance(
    cell95_remaining_pilot_queue,
    list,
):

    for index, item in enumerate(
        cell95_remaining_pilot_queue
    ):

        contains_current = any(
            scalar
            ==
            cell106_current_pilot_id

            for _, scalar
            in cell106_walk(
                item
            )
        )


        if (
            contains_current
        ):

            cell106_queue_entries_with_current_pilot.append(
                {
                    "index":
                        index,

                    "item":
                        item,
                }
            )


cell106_next_queue_preview = []


if (
    isinstance(
        cell95_remaining_pilot_queue,
        list,
    )

    and

    len(
        cell95_remaining_pilot_queue
    )
    >
    0
):

    for index, item in enumerate(
        cell95_remaining_pilot_queue[
            :3
        ]
    ):

        cell106_next_queue_preview.append(
            {
                "index":
                    index,

                "item":
                    item,
            }
        )


# ============================================================
# 7) Conservative transition interpretation
# ============================================================

cell106_explicit_reject_cues = [
    item

    for item
    in cell106_contract_transition_hits

    if any(
        term
        in
        item[
            "matched_terms"
        ]

        for term
        in [
            "REJECT",
            "REJECTED",
        ]
    )
]


cell106_explicit_revision_cues = [
    item

    for item
    in cell106_contract_transition_hits

    if (
        "REVISION_REQUIRED"
        in
        item[
            "matched_terms"
        ]
    )
]


cell106_explicit_queue_cues = [
    item

    for item
    in cell106_contract_transition_hits

    if any(
        term
        in
        item[
            "matched_terms"
        ]

        for term
        in [
            "QUEUE",
            "NEXT",
            "ADVANCE",
            "SKIP",
            "SKIPPED",
        ]
    )
]


cell106_explicit_repair_cues = [
    item

    for item
    in cell106_contract_transition_hits

    if any(
        term
        in
        item[
            "matched_terms"
        ]

        for term
        in [
            "REPAIR",
            "RETRY",
            "RETARGET",
        ]
    )
]


cell106_has_explicit_post_revision_transition = (
    len(
        cell106_explicit_revision_cues
    )
    >
    0

    and

    (
        len(
            cell106_explicit_reject_cues
        )
        >
        0

        or

        len(
            cell106_explicit_queue_cues
        )
        >
        0

        or

        len(
            cell106_explicit_repair_cues
        )
        >
        0
    )
)


# ============================================================
# 8) Inspection result
# ============================================================

cell106_workflow_inspection_result = {
    "inspection_version":
        "post_revision_workflow_transition_inspection_v1",

    "pilot_case_id":
        cell106_current_pilot_id,

    "current_human_decision":
        cell100_human_adjudication_record[
            "human_decision"
        ],

    "retarget_discovery_result":
        cell105_discovery_result[
            "next_action"
        ],

    "contract_top_level_keys":
        cell106_contract_top_level_keys,

    "transition_hit_count":
        len(
            cell106_contract_transition_hits
        ),

    "lifecycle_candidate_count":
        len(
            cell106_lifecycle_candidates
        ),

    "explicit_reject_cue_count":
        len(
            cell106_explicit_reject_cues
        ),

    "explicit_revision_cue_count":
        len(
            cell106_explicit_revision_cues
        ),

    "explicit_queue_cue_count":
        len(
            cell106_explicit_queue_cues
        ),

    "explicit_repair_cue_count":
        len(
            cell106_explicit_repair_cues
        ),

    "has_explicit_post_revision_transition":
        cell106_has_explicit_post_revision_transition,

    "remaining_queue_count":
        cell106_queue_count,

    "current_pilot_queue_match_count":
        len(
            cell106_queue_entries_with_current_pilot
        ),

    "queue_mutated":
        False,

    "pilot_mutated":
        False,

    "expected_answer_generated":
        False,

    "next_action":
        (
            "INTERPRET_EXPLICIT_WORKFLOW_TRANSITION_CUES"

            if (
                cell106_has_explicit_post_revision_transition
            )

            else
            "NO_EXPLICIT_TRANSITION_CONFIRMED_DO_NOT_ADVANCE_QUEUE"
        ),
}


cell106_workflow_inspection_sha256 = (
    cell106_sha(
        cell106_workflow_inspection_result
    )
)


# ============================================================
# 9) Freeze protected runtime AFTER inspection
# ============================================================

cell106_contract_snapshot_after = (
    cell106_sha(
        PILOT_QA_WORKFLOW_CONTRACT_V1
    )
)


cell106_queue_snapshot_after = (
    cell106_sha(
        cell95_remaining_pilot_queue
    )
)


cell106_pilot_runtime_snapshot_after = (
    cell106_sha(
        {
            "question_drafting_pilot_packets":
                question_drafting_pilot_packets,

            "question_drafting_pilot_by_id":
                question_drafting_pilot_by_id,
        }
    )
)


cell106_adjudication_snapshot_after = (
    cell106_sha(
        cell100_human_adjudication_record
    )
)


cell106_repair_snapshot_after = (
    cell106_sha(
        cell104_repair_route
    )
)


cell106_discovery_snapshot_after = (
    cell106_sha(
        cell105_discovery_result
    )
)


# ============================================================
# 10) Final gates
# ============================================================

cell106_gates = {
    "cell105_is_pass":
        (
            globals().get(
                "cell_105_pass"
            )
            is True
        ),

    "retarget_discovery_found_no_existing_candidate":
        (
            cell105_discovery_result.get(
                "eligible_alternative_candidate_count"
            )
            ==
            0
        ),

    "workflow_contract_was_inspected":
        (
            len(
                cell106_contract_top_level_keys
            )
            >
            0
        ),

    "transition_cues_were_collected":
        (
            isinstance(
                cell106_contract_transition_hits,
                list,
            )
        ),

    "queue_was_inspected":
        (
            cell106_queue_count
            is not None
        ),

    "queue_not_mutated":
        (
            cell106_workflow_inspection_result[
                "queue_mutated"
            ]
            is False
        ),

    "pilot_not_mutated":
        (
            cell106_workflow_inspection_result[
                "pilot_mutated"
            ]
            is False
        ),

    "expected_answer_not_generated":
        (
            cell106_workflow_inspection_result[
                "expected_answer_generated"
            ]
            is False
        ),

    "workflow_contract_remains_byte_stable":
        (
            cell106_contract_snapshot_before
            ==
            cell106_contract_snapshot_after
        ),

    "remaining_queue_remains_byte_stable":
        (
            cell106_queue_snapshot_before
            ==
            cell106_queue_snapshot_after
        ),

    "pilot_runtime_remains_byte_stable":
        (
            cell106_pilot_runtime_snapshot_before
            ==
            cell106_pilot_runtime_snapshot_after
        ),

    "human_adjudication_remains_byte_stable":
        (
            cell106_adjudication_snapshot_before
            ==
            cell106_adjudication_snapshot_after
        ),

    "repair_route_remains_byte_stable":
        (
            cell106_repair_snapshot_before
            ==
            cell106_repair_snapshot_after
        ),

    "discovery_result_remains_byte_stable":
        (
            cell106_discovery_snapshot_before
            ==
            cell106_discovery_snapshot_after
        ),

    "no_artifact_write":
        True,
}


cell_106_pass = all(
    cell106_gates.values()
)


cell106_status = (
    "POST_REVISION_WORKFLOW_CONTRACT_INSPECTED"

    if (
        cell_106_pass
    )

    else
    "POST_REVISION_WORKFLOW_CONTRACT_INSPECTION_FAILED"
)


# ============================================================
# 11) Diagnostics
# ============================================================

print(
    "=== POST-REVISION WORKFLOW CONTRACT INSPECTION ==="
)


print(
    "\n=== CURRENT STATE ==="
)

print(
    "Pilot ID:",
    cell106_current_pilot_id
)

print(
    "Human decision:",
    cell100_human_adjudication_record[
        "human_decision"
    ]
)

print(
    "Retarget discovery:",
    cell105_discovery_result[
        "next_action"
    ]
)

print(
    "Remaining queue count:",
    cell106_queue_count
)

print(
    "Current Pilot queue matches:",
    len(
        cell106_queue_entries_with_current_pilot
    )
)


print(
    "\n=== WORKFLOW CONTRACT TOP-LEVEL KEYS ==="
)

print(
    cell106_contract_top_level_keys
)


print(
    "\n=== LIFECYCLE / STAGE CANDIDATES ==="
)


for item in (
    cell106_lifecycle_candidates
):

    print(
        "-",
        item[
            "path"
        ],
        "=",
        item[
            "value"
        ]
    )


print(
    "\n=== TRANSITION CUES ==="
)


if (
    cell106_contract_transition_hits
):

    for index, item in enumerate(
        cell106_contract_transition_hits,
        start=1,
    ):

        print(
            f"\n--- Cue {index} ---"
        )

        print(
            "Path:",
            item[
                "path"
            ]
        )

        print(
            "Value:",
            item[
                "value"
            ]
        )

        print(
            "Matched terms:",
            item[
                "matched_terms"
            ]
        )


else:

    print(
        "No transition cues found."
    )


print(
    "\n=== CUE COUNTS ==="
)

print(
    "Reject cues:",
    len(
        cell106_explicit_reject_cues
    )
)

print(
    "Revision cues:",
    len(
        cell106_explicit_revision_cues
    )
)

print(
    "Queue cues:",
    len(
        cell106_explicit_queue_cues
    )
)

print(
    "Repair cues:",
    len(
        cell106_explicit_repair_cues
    )
)

print(
    "Explicit post-revision transition available:",
    cell106_has_explicit_post_revision_transition
)


print(
    "\n=== QUEUE PREVIEW ==="
)


for item in (
    cell106_next_queue_preview
):

    print(
        "\nQueue index:",
        item[
            "index"
        ]
    )

    print(
        item[
            "item"
        ]
    )


print(
    "\nNext action:",
    cell106_workflow_inspection_result[
        "next_action"
    ]
)

print(
    "Inspection SHA256:",
    cell106_workflow_inspection_sha256
)


print(
    "\n=== FAILURES ==="
)

print(
    "Prerequisite failures:",
    cell106_prerequisite_failures
)


print(
    "\nSTATUS:",
    cell106_status
)


print(
    "\n=== CODE CELL 106 GATES ==="
)


for gate_name, gate_value in (
    cell106_gates.items()
):

    print(
        gate_name,
        ":",
        (
            "PASS"
            if gate_value
            else "FAIL"
        )
    )


print(
    "\nFAILED GATES:",
    [
        gate_name

        for gate_name, gate_value
        in cell106_gates.items()

        if not gate_value
    ]
)


print(
    "\nCODE CELL 106 GATE:",
    (
        "PASS"
        if cell_106_pass
        else "FAIL"
    )
)


assert cell_106_pass, (
    "Post-revision workflow contract inspection failed. "
    "Do not advance the queue or modify the Pilot. "
    "Repair this SAME Code Cell 106 using the printed diagnostics."
)
```

<h3>بررسی مستقیم Question Revision Policy</h3>

<p>
Cell 106 نشان داد Workflow Contract یک بخش مستقل با نام
<code>question_revision_policy</code>
دارد.
از آنجا که Transition پس از
<code>REVISION_REQUIRED</code>

<p>
هدف این است که مشخص شود Contract درباره موارد زیر چه می‌گوید:
</p>

<ul>

<li>مجاز یا غیرمجاز بودن Queue advancement؛</li>
<li>مجاز یا غیرمجاز بودن ساخت Expected Answer پیش از Question Acceptance.</li>
</ul>

<p>
این Cell فقط Read-Only است و هیچ Question، Evidence، Queue، Pilot یا Artifact را تغییر نمی‌دهد.
</p>

<h4>معیار PASS</h4>

<p>
محتوای واقعی <code>question_revision_policy</code> و تمام scalarهای مرتبط با آن باید چاپ شوند.
اگر Policy مسیر دقیقی برای وضعیت فعلی نداشته باشد، همان نبود Rule صریح ثبت می‌شود و هیچ Transition جدیدی ساخته نمی‌شود.
</p>

```python
# Code Cell 107

# ============================================================
# Direct read-only inspection of question_revision_policy.
#
# No question revision.
# No evidence retargeting.
# No queue advancement.
# No expected-answer generation.
# No artifact write.
# ============================================================


# ============================================================
# 1) Prerequisites
# ============================================================

cell107_prerequisite_failures = []


if (
    globals().get(
        "cell_106_pass"
    )
    is not True
):

    cell107_prerequisite_failures.append(
        "CELL106_NOT_PASS"
    )


if (
    globals().get(
        "cell106_status"
    )
    !=
    "POST_REVISION_WORKFLOW_CONTRACT_INSPECTED"
):

    cell107_prerequisite_failures.append(
        {
            "reason":
                "CELL106_STATUS_UNEXPECTED",

            "actual":
                globals().get(
                    "cell106_status"
                ),
        }
    )


CELL107_REQUIRED_RUNTIME_OBJECTS = [
    "PILOT_QA_WORKFLOW_CONTRACT_V1",
    "cell106_workflow_inspection_result",
    "cell100_human_adjudication_record",
    "cell104_repair_route",
    "cell105_discovery_result",
    "cell95_remaining_pilot_queue",
]


cell107_missing_runtime_objects = [
    name

    for name
    in CELL107_REQUIRED_RUNTIME_OBJECTS

    if (
        name
        not in
        globals()
    )
]


if (
    cell107_missing_runtime_objects
):

    cell107_prerequisite_failures.append(
        {
            "reason":
                "REQUIRED_RUNTIME_OBJECTS_MISSING",

            "objects":
                cell107_missing_runtime_objects,
        }
    )


# ============================================================
# 2) Helpers
# ============================================================

def cell107_sha(value):

    return hashlib.sha256(
        stable_json_key(
            value
        ).encode(
            "utf-8"
        )
    ).hexdigest()


def cell107_walk(value, path=()):

    if isinstance(
        value,
        dict,
    ):

        for key, child in value.items():

            yield from cell107_walk(
                child,
                path + (str(key),),
            )

    elif isinstance(
        value,
        (list, tuple),
    ):

        for index, child in enumerate(
            value
        ):

            yield from cell107_walk(
                child,
                path + (str(index),),
            )

    else:

        yield path, value


# ============================================================
# 3) Freeze protected state BEFORE inspection
# ============================================================

cell107_contract_snapshot_before = (
    cell107_sha(
        PILOT_QA_WORKFLOW_CONTRACT_V1
    )
)


cell107_queue_snapshot_before = (
    cell107_sha(
        cell95_remaining_pilot_queue
    )
)


cell107_adjudication_snapshot_before = (
    cell107_sha(
        cell100_human_adjudication_record
    )
)


cell107_repair_snapshot_before = (
    cell107_sha(
        cell104_repair_route
    )
)


cell107_discovery_snapshot_before = (
    cell107_sha(
        cell105_discovery_result
    )
)


cell107_pilot_snapshot_before = (
    cell107_sha(
        {
            "question_drafting_pilot_packets":
                question_drafting_pilot_packets,

            "question_drafting_pilot_by_id":
                question_drafting_pilot_by_id,
        }
    )
)


# ============================================================
# 4) Resolve exact question revision policy
# ============================================================

cell107_question_revision_policy = (
    PILOT_QA_WORKFLOW_CONTRACT_V1.get(
        "question_revision_policy"
    )

    if isinstance(
        PILOT_QA_WORKFLOW_CONTRACT_V1,
        dict,
    )

    else None
)


cell107_policy_failures = []


if (
    cell107_question_revision_policy
    is None
):

    cell107_policy_failures.append(
        "QUESTION_REVISION_POLICY_MISSING"
    )


# ============================================================
# 5) Flatten policy for exact inspection
# ============================================================

cell107_policy_scalars = []


if (
    cell107_question_revision_policy
    is not None
):

    for path, scalar in cell107_walk(
        cell107_question_revision_policy
    ):

        cell107_policy_scalars.append(
            {
                "path":
                    ".".join(
                        path
                    ),

                "value":
                    scalar,
            }
        )


# ============================================================
# 6) Extract explicit semantic cues
# ============================================================

CELL107_CUE_TERMS = [
    "REVISION_REQUIRED",
    "REVISE",
    "REVISION",
    "QUESTION",
    "EVIDENCE",
    "TARGET",
    "MACHINE",
    "HUMAN",
    "ACCEPT",
    "REJECT",
    "BLOCKED",
    "EXPECTED_ANSWER",
    "QUEUE",
    "NEXT",
    "ADVANCE",
    "REVALIDATE",
    "VALIDATE",
]


cell107_policy_cues = []


for item in (
    cell107_policy_scalars
):

    path_text = (
        item[
            "path"
        ]
    )

    value_text = (
        str(
            item[
                "value"
            ]
        )
    )


    matched_terms = [
        term

        for term
        in CELL107_CUE_TERMS

        if (
            term.lower()
            in
            path_text.lower()

            or

            term.lower()
            in
            value_text.lower()
        )
    ]


    if (
        matched_terms
    ):

        cell107_policy_cues.append(
            {
                "path":
                    path_text,

                "value":
                    item[
                        "value"
                    ],

                "matched_terms":
                    matched_terms,
            }
        )


# ============================================================
# 7) Conservative policy interpretation
# ============================================================

cell107_mentions_revision_required = any(
    "REVISION_REQUIRED"
    in
    item[
        "matched_terms"
    ]

    for item
    in cell107_policy_cues
)


cell107_mentions_question_revision = any(
    (
        "QUESTION"
        in
        item[
            "matched_terms"
        ]

        and

        (
            "REVISION"
            in
            item[
                "matched_terms"
            ]

            or

            "REVISE"
            in
            item[
                "matched_terms"
            ]
        )
    )

    for item
    in cell107_policy_cues
)


cell107_mentions_evidence_or_target_revision = any(
    (
        "EVIDENCE"
        in
        item[
            "matched_terms"
        ]

        or

        "TARGET"
        in
        item[
            "matched_terms"
        ]
    )

    for item
    in cell107_policy_cues
)


cell107_mentions_machine_revalidation = any(
    (
        "MACHINE"
        in
        item[
            "matched_terms"
        ]

        and

        (
            "VALIDATE"
            in
            item[
                "matched_terms"
            ]

            or

            "REVALIDATE"
            in
            item[
                "matched_terms"
            ]
        )
    )

    for item
    in cell107_policy_cues
)


cell107_mentions_queue_advancement = any(
    (
        "QUEUE"
        in
        item[
            "matched_terms"
        ]

        or

        "ADVANCE"
        in
        item[
            "matched_terms"
        ]

        or

        "NEXT"
        in
        item[
            "matched_terms"
        ]
    )

    for item
    in cell107_policy_cues
)


cell107_mentions_expected_answer_block = any(
    (
        "EXPECTED_ANSWER"
        in
        item[
            "matched_terms"
        ]

        and

        "BLOCKED"
        in
        item[
            "matched_terms"
        ]
    )

    for item
    in cell107_policy_cues
)


# ============================================================
# 8) Determine whether current repair route is contract-backed
# ============================================================

cell107_current_repair_route = (
    cell104_repair_route.get(
        "recommended_repair_action"
    )
)


cell107_retarget_explicitly_supported_by_policy = any(
    (
        "TARGET"
        in
        item[
            "matched_terms"
        ]

        or

        "EVIDENCE"
        in
        item[
            "matched_terms"
        ]
    )

    and

    (
        "REVISION"
        in
        item[
            "matched_terms"
        ]

        or

        "REVISE"
        in
        item[
            "matched_terms"
        ]
    )

    for item
    in cell107_policy_cues
)


cell107_question_only_revision_policy = (
    cell107_mentions_question_revision
    and
    not cell107_mentions_evidence_or_target_revision
)


# ============================================================
# 9) Inspection result
# ============================================================

cell107_revision_policy_inspection = {
    "inspection_version":
        "question_revision_policy_direct_inspection_v1",

    "pilot_case_id":
        cell100_human_adjudication_record[
            "pilot_case_id"
        ],

    "human_decision":
        cell100_human_adjudication_record[
            "human_decision"
        ],

    "current_repair_route":
        cell107_current_repair_route,

    "policy_present":
        (
            cell107_question_revision_policy
            is not None
        ),

    "policy_scalar_count":
        len(
            cell107_policy_scalars
        ),

    "policy_cue_count":
        len(
            cell107_policy_cues
        ),

    "mentions_revision_required":
        cell107_mentions_revision_required,

    "mentions_question_revision":
        cell107_mentions_question_revision,

    "mentions_evidence_or_target_revision":
        cell107_mentions_evidence_or_target_revision,

    "mentions_machine_revalidation":
        cell107_mentions_machine_revalidation,

    "mentions_queue_advancement":
        cell107_mentions_queue_advancement,

    "mentions_expected_answer_block":
        cell107_mentions_expected_answer_block,

    "retarget_explicitly_supported_by_policy":
        cell107_retarget_explicitly_supported_by_policy,

    "question_only_revision_policy":
        cell107_question_only_revision_policy,

    "queue_mutated":
        False,

    "pilot_mutated":
        False,

    "question_mutated":
        False,

    "evidence_mutated":
        False,

    "expected_answer_generated":
        False,
}


if (
    cell107_question_only_revision_policy
):

    cell107_revision_policy_inspection[
        "next_action"
    ] = (
        "CURRENT_RETARGET_ROUTE_NOT_CONTRACT_BACKED_"
        "REASSESS_QUESTION_REVISION_PATH"
    )


elif (
    cell107_retarget_explicitly_supported_by_policy
):

    cell107_revision_policy_inspection[
        "next_action"
    ] = (
        "RETARGET_ROUTE_HAS_POLICY_SUPPORT_"
        "BUT_NO_VALIDATED_ALTERNATIVE_EXISTS_KEEP_BLOCKED"
    )


else:

    cell107_revision_policy_inspection[
        "next_action"
    ] = (
        "POLICY_DOES_NOT_EXPLICITLY_RESOLVE_CURRENT_STATE_"
        "DO_NOT_ADVANCE_OR_GENERATE"
    )


cell107_revision_policy_inspection_sha256 = (
    cell107_sha(
        cell107_revision_policy_inspection
    )
)


# ============================================================
# 10) Freeze protected state AFTER inspection
# ============================================================

cell107_contract_snapshot_after = (
    cell107_sha(
        PILOT_QA_WORKFLOW_CONTRACT_V1
    )
)


cell107_queue_snapshot_after = (
    cell107_sha(
        cell95_remaining_pilot_queue
    )
)


cell107_adjudication_snapshot_after = (
    cell107_sha(
        cell100_human_adjudication_record
    )
)


cell107_repair_snapshot_after = (
    cell107_sha(
        cell104_repair_route
    )
)


cell107_discovery_snapshot_after = (
    cell107_sha(
        cell105_discovery_result
    )
)


cell107_pilot_snapshot_after = (
    cell107_sha(
        {
            "question_drafting_pilot_packets":
                question_drafting_pilot_packets,

            "question_drafting_pilot_by_id":
                question_drafting_pilot_by_id,
        }
    )
)


# ============================================================
# 11) Final gates
# ============================================================

cell107_gates = {
    "cell106_is_pass":
        (
            globals().get(
                "cell_106_pass"
            )
            is True
        ),

    "question_revision_policy_exists":
        (
            len(
                cell107_policy_failures
            )
            ==
            0
        ),

    "policy_was_directly_inspected":
        (
            len(
                cell107_policy_scalars
            )
            >
            0
        ),

    "queue_not_mutated":
        (
            cell107_revision_policy_inspection[
                "queue_mutated"
            ]
            is False
        ),

    "pilot_not_mutated":
        (
            cell107_revision_policy_inspection[
                "pilot_mutated"
            ]
            is False
        ),

    "question_not_mutated":
        (
            cell107_revision_policy_inspection[
                "question_mutated"
            ]
            is False
        ),

    "evidence_not_mutated":
        (
            cell107_revision_policy_inspection[
                "evidence_mutated"
            ]
            is False
        ),

    "expected_answer_not_generated":
        (
            cell107_revision_policy_inspection[
                "expected_answer_generated"
            ]
            is False
        ),

    "contract_remains_byte_stable":
        (
            cell107_contract_snapshot_before
            ==
            cell107_contract_snapshot_after
        ),

    "queue_remains_byte_stable":
        (
            cell107_queue_snapshot_before
            ==
            cell107_queue_snapshot_after
        ),

    "adjudication_remains_byte_stable":
        (
            cell107_adjudication_snapshot_before
            ==
            cell107_adjudication_snapshot_after
        ),

    "repair_route_remains_byte_stable":
        (
            cell107_repair_snapshot_before
            ==
            cell107_repair_snapshot_after
        ),

    "discovery_result_remains_byte_stable":
        (
            cell107_discovery_snapshot_before
            ==
            cell107_discovery_snapshot_after
        ),

    "pilot_runtime_remains_byte_stable":
        (
            cell107_pilot_snapshot_before
            ==
            cell107_pilot_snapshot_after
        ),

    "no_artifact_write":
        True,
}


cell_107_pass = all(
    cell107_gates.values()
)


cell107_status = (
    "QUESTION_REVISION_POLICY_INSPECTED"

    if (
        cell_107_pass
    )

    else
    "QUESTION_REVISION_POLICY_INSPECTION_FAILED"
)


# ============================================================
# 12) Diagnostics
# ============================================================

print(
    "=== QUESTION REVISION POLICY — DIRECT INSPECTION ==="
)


print(
    "\n=== RAW POLICY ==="
)

print(
    cell107_question_revision_policy
)


print(
    "\n=== POLICY SCALARS ==="
)


for item in (
    cell107_policy_scalars
):

    print(
        "-",
        item[
            "path"
        ],
        "=",
        item[
            "value"
        ]
    )


print(
    "\n=== POLICY CUES ==="
)


if (
    cell107_policy_cues
):

    for index, item in enumerate(
        cell107_policy_cues,
        start=1,
    ):

        print(
            f"\n--- Cue {index} ---"
        )

        print(
            "Path:",
            item[
                "path"
            ]
        )

        print(
            "Value:",
            item[
                "value"
            ]
        )

        print(
            "Matched terms:",
            item[
                "matched_terms"
            ]
        )


else:

    print(
        "No policy cues found."
    )


print(
    "\n=== INTERPRETATION ==="
)

print(
    "Mentions REVISION_REQUIRED:",
    cell107_mentions_revision_required
)

print(
    "Mentions Question revision:",
    cell107_mentions_question_revision
)

print(
    "Mentions Evidence/Target revision:",
    cell107_mentions_evidence_or_target_revision
)

print(
    "Mentions Machine revalidation:",
    cell107_mentions_machine_revalidation
)

print(
    "Mentions Queue advancement:",
    cell107_mentions_queue_advancement
)

print(
    "Mentions Expected Answer block:",
    cell107_mentions_expected_answer_block
)

print(
    "Retarget explicitly supported:",
    cell107_retarget_explicitly_supported_by_policy
)

print(
    "Question-only revision policy:",
    cell107_question_only_revision_policy
)

print(
    "Next action:",
    cell107_revision_policy_inspection[
        "next_action"
    ]
)

print(
    "Inspection SHA256:",
    cell107_revision_policy_inspection_sha256
)


print(
    "\n=== FAILURES ==="
)

print(
    "Prerequisite failures:",
    cell107_prerequisite_failures
)

print(
    "Policy failures:",
    cell107_policy_failures
)


print(
    "\nSTATUS:",
    cell107_status
)


print(
    "\n=== CODE CELL 107 GATES ==="
)


for gate_name, gate_value in (
    cell107_gates.items()
):

    print(
        gate_name,
        ":",
        (
            "PASS"
            if gate_value
            else "FAIL"
        )
    )


print(
    "\nFAILED GATES:",
    [
        gate_name

        for gate_name, gate_value
        in cell107_gates.items()

        if not gate_value
    ]
)


print(
    "\nCODE CELL 107 GATE:",
    (
        "PASS"
        if cell_107_pass
        else "FAIL"
    )
)


assert cell_107_pass, (
    "Question revision policy inspection failed. "
    "Do not revise the Question, retarget Evidence, "
    "advance the queue, or generate an Expected Answer. "
    "Repair this SAME Code Cell 107 using the printed diagnostics."
)
```

<h3>بررسی Schema واقعی Revision ID و Lineage</h3>

<p>
Cell 107 نشان داد تنها Rule صریح Contract برای Revision این است:
<code>CREATE_NEW_REVISION_ID_PRESERVE_LINEAGE</code>.
</p>

<p>
این Rule مشخص نمی‌کند Revision Record دقیقاً چه Fieldهایی دارد.
برای جلوگیری از ساخت Schema حدسی، این مرحله فقط Runtimeهای موجود را Read-Only بررسی می‌کند تا هر Field یا Pattern واقعی مرتبط با Revision و Lineage پیدا شود.
</p>

<p>
موارد مورد بررسی:
</p>

<ul>
<li><code>question_draft_id</code> و هر Revision-related ID؛</li>
<li>Fieldهای حاوی <code>revision</code>، <code>lineage</code>، <code>parent</code>، <code>supersede</code> یا <code>previous</code>؛</li>
<li>ساختار Question Draft فعلی؛</li>
<li>ساختار Reference Completed Pilot؛</li>
<li>هر Runtime دیگری که Schema واقعی Revision را اثبات کند.</li>
</ul>

<p>
هیچ Revision جدیدی ساخته نمی‌شود، هیچ Question یا Evidence تغییر نمی‌کند و Queue نیز دست‌نخورده باقی می‌ماند.
</p>

<h4>معیار PASS</h4>

<p>
تمام Cueهای واقعی Revision/Lineage از Runtime استخراج و چاپ شوند.
اگر Schema کامل Revision در Runtime موجود نباشد، نتیجه باید همین محدودیت را گزارش کند و هیچ Field جدیدی حدس زده نشود.
</p>

</div>

<h3>تفکیک Revision Lineage واقعی از False Positiveهای Cell 108</h3>

<p>
Cell 108 از نظر اجرای فنی PASS بود، اما نتیجه
<code>has_explicit_lineage_schema = True</code>
باید دقیق‌تر بررسی شود.
</p>

<p>
علت این است که چند Field شامل واژه <code>parent</code> مربوط به Task Family
<code>PARENT_SUPPLEMENTAL_EVIDENCE_RETRIEVAL</code>
هستند و لزوماً هیچ ارتباطی با Parent/Lineage یک Question Revision ندارند.
</p>

<p>
این مرحله دو کار Read-Only انجام می‌دهد:
</p>

<ul>
<li>Revision/Lineage cueهای Cell 108 را به Policy، Human Adjudication، Task-semantic Parent و Structural Revision Linkage تفکیک می‌کند؛</li>
<li>History واقعی همین Notebook را برای Implementation قبلی ساخت Revision، Revision ID یا Lineage بررسی می‌کند.</li>
</ul>

<p>
وجود واژه‌هایی مانند <code>revision</code> یا <code>parent</code> به‌تنهایی Schema محسوب نمی‌شود.
برای اثبات Revision Schema باید یک Field یا Code Pattern واقعی برای اتصال Revision جدید به Question Draft قبلی وجود داشته باشد.
</p>

<p>
در این Cell هیچ Revision ساخته نمی‌شود، هیچ تصمیم انسانی تغییر نمی‌کند، هیچ Question یا Evidence اصلاح نمی‌شود و Queue نیز دست‌نخورده باقی می‌ماند.
</p>

<h4>معیار PASS</h4>

<p>
False Positiveهای Parent از Revision Lineage واقعی جدا شوند و Notebook History بدون Mutation بررسی شود.
اگر Implementation واقعی Revision پیدا نشود، نتیجه باید صریحاً
<code>NO_CONCRETE_REVISION_LINKAGE_IMPLEMENTATION_FOUND</code>

</p>

```python
# Code Cell 109

# ============================================================
# Refine Revision / Lineage interpretation.
#
# Distinguish:
# - workflow policy cues
# - adjudication-state cues
# - task-semantic parent cues
# - concrete revision-linkage fields / implementation
#
# Also inspect Notebook input history read-only.
#
# No revision creation.
# No question mutation.
# No evidence mutation.
# No queue mutation.
# No expected-answer generation.
# No artifact write.
# ============================================================


# ============================================================
# 1) Prerequisites
# ============================================================

cell109_prerequisite_failures = []


if (
    globals().get(
        "cell_108_pass"
    )
    is not True
):

    cell109_prerequisite_failures.append(
        "CELL108_NOT_PASS"
    )


if (
    globals().get(
        "cell108_status"
    )
    !=
    "REVISION_LINEAGE_SCHEMA_INSPECTED"
):

    cell109_prerequisite_failures.append(
        {
            "reason":
                "CELL108_STATUS_UNEXPECTED",

            "actual":
                globals().get(
                    "cell108_status"
                ),
        }
    )


CELL109_REQUIRED_RUNTIME_OBJECTS = [
    "cell108_revision_lineage_cues",
    "cell108_revision_schema_inspection",
    "cell97_question_draft_record",
    "cell100_human_adjudication_record",
    "cell104_repair_route",
    "cell95_remaining_pilot_queue",
    "question_drafting_pilot_packets",
    "question_drafting_pilot_by_id",
    "PILOT_QA_WORKFLOW_CONTRACT_V1",
]


cell109_missing_runtime_objects = [
    name

    for name
    in CELL109_REQUIRED_RUNTIME_OBJECTS

    if (
        name
        not in
        globals()
    )
]


if (
    cell109_missing_runtime_objects
):

    cell109_prerequisite_failures.append(
        {
            "reason":
                "REQUIRED_RUNTIME_OBJECTS_MISSING",

            "objects":
                cell109_missing_runtime_objects,
        }
    )


# ============================================================
# 2) Helpers
# ============================================================

def cell109_sha(value):

    return hashlib.sha256(
        stable_json_key(
            value
        ).encode(
            "utf-8"
        )
    ).hexdigest()


def cell109_walk(value, path=()):

    if isinstance(
        value,
        dict,
    ):

        for key, child in value.items():

            yield from cell109_walk(
                child,
                path + (str(key),),
            )

    elif isinstance(
        value,
        (list, tuple),
    ):

        for index, child in enumerate(
            value
        ):

            yield from cell109_walk(
                child,
                path + (str(index),),
            )

    else:

        yield path, value


# ============================================================
# 3) Freeze protected state BEFORE inspection
# ============================================================

cell109_question_snapshot_before = (
    cell109_sha(
        cell97_question_draft_record
    )
)


cell109_adjudication_snapshot_before = (
    cell109_sha(
        cell100_human_adjudication_record
    )
)


cell109_repair_snapshot_before = (
    cell109_sha(
        cell104_repair_route
    )
)


cell109_queue_snapshot_before = (
    cell109_sha(
        cell95_remaining_pilot_queue
    )
)


cell109_pilot_snapshot_before = (
    cell109_sha(
        {
            "question_drafting_pilot_packets":
                question_drafting_pilot_packets,

            "question_drafting_pilot_by_id":
                question_drafting_pilot_by_id,
        }
    )
)


cell109_contract_snapshot_before = (
    cell109_sha(
        PILOT_QA_WORKFLOW_CONTRACT_V1
    )
)


# ============================================================
# 4) Define search patterns only
#
# These are inspection patterns, NOT invented schema fields.
# ============================================================

CELL109_CONCRETE_REVISION_LINK_PATTERNS = [
    "revision_id",
    "revision_of",
    "revision_parent",
    "parent_revision",
    "parent_question",
    "previous_question",
    "prior_question",
    "supersedes",
    "superseded_by",
    "replaces_question",
    "replaced_by",
    "lineage_parent",
    "lineage_id",
    "revision_number",
]


CELL109_TASK_PARENT_SEMANTIC_MARKERS = [
    "PARENT_SUPPLEMENTAL",
    "PARENT_CONTEXT",
    "PARENT_PLUS_SUPPLEMENTAL",
    "USE_PARENT_PLUS_SUPPLEMENTAL",
    "EXACT_PARENT_SUPPLEMENTAL",
    "PARENT_PARAGRAPH",
    "parent_paragraph_id",
]


# ============================================================
# 5) Reclassify Cell 108 cues
# ============================================================

cell109_policy_cues = []
cell109_adjudication_revision_cues = []
cell109_task_parent_semantic_cues = []
cell109_concrete_runtime_revision_linkage_cues = []
cell109_other_cues = []


for item in (
    cell108_revision_lineage_cues
):

    root = (
        item.get(
            "root",
            "",
        )
    )


    path = (
        item.get(
            "path",
            ""
        )
    )


    value = (
        item.get(
            "value"
        )
    )


    combined = (
        f"{path}\n{value}"
    ).lower()


    is_policy = (
        root
        ==
        "PILOT_QA_WORKFLOW_CONTRACT_V1"

        and

        path
        ==
        "question_revision_policy"
    )


    is_adjudication_revision_state = (
        root
        ==
        "cell100_human_adjudication_record"

        and

        path
        in
        [
            "human_decision",
            "revision_type",
        ]
    )


    is_task_parent_semantic = any(
        marker.lower()
        in
        combined

        for marker
        in CELL109_TASK_PARENT_SEMANTIC_MARKERS
    )


    is_concrete_runtime_revision_linkage = any(
        pattern
        in
        path.lower()

        for pattern
        in CELL109_CONCRETE_REVISION_LINK_PATTERNS
    )


    classified = {
        "root":
            root,

        "path":
            path,

        "value":
            value,
    }


    if (
        is_policy
    ):

        cell109_policy_cues.append(
            classified
        )


    elif (
        is_adjudication_revision_state
    ):

        cell109_adjudication_revision_cues.append(
            classified
        )


    elif (
        is_task_parent_semantic
    ):

        cell109_task_parent_semantic_cues.append(
            classified
        )


    elif (
        is_concrete_runtime_revision_linkage
    ):

        cell109_concrete_runtime_revision_linkage_cues.append(
            classified
        )


    else:

        cell109_other_cues.append(
            classified
        )


# ============================================================
# 6) Search all relevant runtime objects for concrete linkage
# ============================================================

cell109_runtime_roots = {
    "cell97_question_draft_record":
        cell97_question_draft_record,

    "cell100_human_adjudication_record":
        cell100_human_adjudication_record,

    "cell104_repair_route":
        cell104_repair_route,

    "question_drafting_pilot_packets":
        question_drafting_pilot_packets,

    "question_drafting_pilot_by_id":
        question_drafting_pilot_by_id,

    "PILOT_QA_WORKFLOW_CONTRACT_V1":
        PILOT_QA_WORKFLOW_CONTRACT_V1,
}


cell109_concrete_runtime_field_hits = []


for root_name, root_value in (
    cell109_runtime_roots.items()
):

    for path, scalar in cell109_walk(
        root_value
    ):

        path_text = (
            ".".join(
                path
            )
        )


        matched_patterns = [
            pattern

            for pattern
            in CELL109_CONCRETE_REVISION_LINK_PATTERNS

            if (
                pattern
                in
                path_text.lower()
            )
        ]


        if (
            matched_patterns
        ):

            cell109_concrete_runtime_field_hits.append(
                {
                    "root":
                        root_name,

                    "path":
                        path_text,

                    "value":
                        scalar,

                    "matched_patterns":
                        matched_patterns,
                }
            )


# ============================================================
# 7) Read Notebook input history
# ============================================================

cell109_history_failures = []


try:

    cell109_ipython = (
        get_ipython()
    )


except Exception as exc:

    cell109_ipython = None

    cell109_history_failures.append(
        {
            "reason":
                "GET_IPYTHON_FAILED",

            "error":
                repr(
                    exc
                ),
        }
    )


cell109_input_history = []


if (
    cell109_ipython
    is not None
):

    try:

        cell109_input_history = list(
            cell109_ipython.history_manager.input_hist_raw
        )

    except Exception as exc:

        cell109_history_failures.append(
            {
                "reason":
                    "INPUT_HISTORY_READ_FAILED",

                "error":
                    repr(
                        exc
                    ),
            }
        )


# ============================================================
# 8) Search Notebook history for revision implementation
# ============================================================

CELL109_HISTORY_SEARCH_TERMS = [
    "CREATE_NEW_REVISION_ID_PRESERVE_LINEAGE",
    "revision_id",
    "revision_of",
    "lineage",
    "supersedes",
    "previous_question",
    "parent_question",
]


cell109_history_term_hits = []


for history_index, source in enumerate(
    cell109_input_history
):

    if not isinstance(
        source,
        str,
    ):

        continue


    matched_terms = [
        term

        for term
        in CELL109_HISTORY_SEARCH_TERMS

        if (
            term.lower()
            in
            source.lower()
        )
    ]


    if not (
        matched_terms
    ):

        continue


    matching_lines = [
        line.strip()

        for line
        in source.splitlines()

        if any(
            term.lower()
            in
            line.lower()

            for term
            in matched_terms
        )
    ]


    cell109_history_term_hits.append(
        {
            "history_index":
                history_index,

            "matched_terms":
                matched_terms,

            "matching_lines":
                matching_lines[
                    :20
                ],
        }
    )


# ============================================================
# 9) Identify actual construction-like history hits
#
# Policy mentions alone are not implementation.
# ============================================================

cell109_history_construction_hits = []


for item in (
    cell109_history_term_hits
):

    lines = (
        item[
            "matching_lines"
        ]
    )


    construction_lines = [
        line

        for line
        in lines

        if (
            "="
            in
            line

            and

            any(
                pattern
                in
                line.lower()

                for pattern
                in CELL109_CONCRETE_REVISION_LINK_PATTERNS
            )
        )
    ]


    if (
        construction_lines
    ):

        cell109_history_construction_hits.append(
            {
                "history_index":
                    item[
                        "history_index"
                    ],

                "construction_lines":
                    construction_lines,
            }
        )


# ============================================================
# 10) Corrected schema interpretation
# ============================================================

cell109_has_concrete_runtime_revision_linkage = (
    len(
        cell109_concrete_runtime_field_hits
    )
    >
    0
)


cell109_has_concrete_history_revision_construction = (
    len(
        cell109_history_construction_hits
    )
    >
    0
)


cell109_has_concrete_revision_linkage_implementation = (
    cell109_has_concrete_runtime_revision_linkage
    or
    cell109_has_concrete_history_revision_construction
)


cell109_cell108_parent_false_positive_confirmed = (
    len(
        cell109_task_parent_semantic_cues
    )
    >
    0

    and

    not (
        any(
            "question"
            in
            item[
                "path"
            ].lower()

            and

            "parent"
            in
            item[
                "path"
            ].lower()

            for item
            in cell109_task_parent_semantic_cues
        )
    )
)


# ============================================================
# 11) Inspection result
# ============================================================

cell109_revision_lineage_interpretation = {
    "inspection_version":
        "revision_lineage_false_positive_refinement_v1",

    "pilot_case_id":
        cell100_human_adjudication_record[
            "pilot_case_id"
        ],

    "question_draft_id":
        cell97_question_draft_record[
            "question_draft_id"
        ],

    "revision_policy":
        cell107_question_revision_policy,

    "policy_cue_count":
        len(
            cell109_policy_cues
        ),

    "adjudication_revision_state_cue_count":
        len(
            cell109_adjudication_revision_cues
        ),

    "task_parent_semantic_cue_count":
        len(
            cell109_task_parent_semantic_cues
        ),

    "concrete_runtime_revision_linkage_field_count":
        len(
            cell109_concrete_runtime_field_hits
        ),

    "notebook_history_search_hit_count":
        len(
            cell109_history_term_hits
        ),

    "notebook_history_construction_hit_count":
        len(
            cell109_history_construction_hits
        ),

    "cell108_parent_false_positive_confirmed":
        cell109_cell108_parent_false_positive_confirmed,

    "has_concrete_runtime_revision_linkage":
        cell109_has_concrete_runtime_revision_linkage,

    "has_concrete_history_revision_construction":
        cell109_has_concrete_history_revision_construction,

    "has_concrete_revision_linkage_implementation":
        cell109_has_concrete_revision_linkage_implementation,

    "revision_created":
        False,

    "question_mutated":
        False,

    "evidence_mutated":
        False,

    "queue_mutated":
        False,

    "pilot_mutated":
        False,

    "expected_answer_generated":
        False,
}


if (
    cell109_has_concrete_revision_linkage_implementation
):

    cell109_revision_lineage_interpretation[
        "schema_status"
    ] = (
        "CONCRETE_REVISION_LINKAGE_IMPLEMENTATION_FOUND"
    )

    cell109_revision_lineage_interpretation[
        "next_action"
    ] = (
        "INSPECT_EXACT_REVISION_CONSTRUCTION_IMPLEMENTATION"
    )


else:

    cell109_revision_lineage_interpretation[
        "schema_status"
    ] = (
        "NO_CONCRETE_REVISION_LINKAGE_IMPLEMENTATION_FOUND"
    )

    cell109_revision_lineage_interpretation[
        "next_action"
    ] = (
        "DO_NOT_INVENT_REVISION_SCHEMA_REASSESS_CURRENT_HUMAN_GATE_STATE"
    )


cell109_revision_lineage_interpretation_sha256 = (
    cell109_sha(
        cell109_revision_lineage_interpretation
    )
)


# ============================================================
# 12) Freeze protected state AFTER inspection
# ============================================================

cell109_question_snapshot_after = (
    cell109_sha(
        cell97_question_draft_record
    )
)


cell109_adjudication_snapshot_after = (
    cell109_sha(
        cell100_human_adjudication_record
    )
)


cell109_repair_snapshot_after = (
    cell109_sha(
        cell104_repair_route
    )
)


cell109_queue_snapshot_after = (
    cell109_sha(
        cell95_remaining_pilot_queue
    )
)


cell109_pilot_snapshot_after = (
    cell109_sha(
        {
            "question_drafting_pilot_packets":
                question_drafting_pilot_packets,

            "question_drafting_pilot_by_id":
                question_drafting_pilot_by_id,
        }
    )
)


cell109_contract_snapshot_after = (
    cell109_sha(
        PILOT_QA_WORKFLOW_CONTRACT_V1
    )
)


# ============================================================
# 13) Final gates
# ============================================================

cell109_gates = {
    "cell108_is_pass":
        (
            globals().get(
                "cell_108_pass"
            )
            is True
        ),

    "revision_policy_is_preserved":
        (
            cell107_question_revision_policy
            ==
            "CREATE_NEW_REVISION_ID_PRESERVE_LINEAGE"
        ),

    "cell108_cues_were_reclassified":
        (
            (
                len(
                    cell109_policy_cues
                )
                +
                len(
                    cell109_adjudication_revision_cues
                )
                +
                len(
                    cell109_task_parent_semantic_cues
                )
                +
                len(
                    cell109_concrete_runtime_revision_linkage_cues
                )
                +
                len(
                    cell109_other_cues
                )
            )
            ==
            len(
                cell108_revision_lineage_cues
            )
        ),

    "notebook_history_was_inspected":
        (
            len(
                cell109_history_failures
            )
            ==
            0
        ),

    "no_revision_created":
        (
            cell109_revision_lineage_interpretation[
                "revision_created"
            ]
            is False
        ),

    "question_not_mutated":
        (
            cell109_revision_lineage_interpretation[
                "question_mutated"
            ]
            is False
        ),

    "evidence_not_mutated":
        (
            cell109_revision_lineage_interpretation[
                "evidence_mutated"
            ]
            is False
        ),

    "queue_not_mutated":
        (
            cell109_revision_lineage_interpretation[
                "queue_mutated"
            ]
            is False
        ),

    "pilot_not_mutated":
        (
            cell109_revision_lineage_interpretation[
                "pilot_mutated"
            ]
            is False
        ),

    "expected_answer_not_generated":
        (
            cell109_revision_lineage_interpretation[
                "expected_answer_generated"
            ]
            is False
        ),

    "question_record_remains_byte_stable":
        (
            cell109_question_snapshot_before
            ==
            cell109_question_snapshot_after
        ),

    "adjudication_record_remains_byte_stable":
        (
            cell109_adjudication_snapshot_before
            ==
            cell109_adjudication_snapshot_after
        ),

    "repair_route_remains_byte_stable":
        (
            cell109_repair_snapshot_before
            ==
            cell109_repair_snapshot_after
        ),

    "remaining_queue_remains_byte_stable":
        (
            cell109_queue_snapshot_before
            ==
            cell109_queue_snapshot_after
        ),

    "pilot_runtime_remains_byte_stable":
        (
            cell109_pilot_snapshot_before
            ==
            cell109_pilot_snapshot_after
        ),

    "workflow_contract_remains_byte_stable":
        (
            cell109_contract_snapshot_before
            ==
            cell109_contract_snapshot_after
        ),

    "no_artifact_write":
        True,
}


cell_109_pass = all(
    cell109_gates.values()
)


cell109_status = (
    "REVISION_LINEAGE_INTERPRETATION_REFINED"

    if (
        cell_109_pass
    )

    else
    "REVISION_LINEAGE_INTERPRETATION_FAILED"
)


# ============================================================
# 14) Diagnostics
# ============================================================

print(
    "=== REVISION / LINEAGE — REFINED INTERPRETATION ==="
)


print(
    "\n=== CELL 108 CUE RECLASSIFICATION ==="
)

print(
    "Policy cues:",
    len(
        cell109_policy_cues
    )
)

print(
    "Adjudication revision-state cues:",
    len(
        cell109_adjudication_revision_cues
    )
)

print(
    "Task-semantic Parent cues:",
    len(
        cell109_task_parent_semantic_cues
    )
)

print(
    "Concrete Runtime revision-linkage cues:",
    len(
        cell109_concrete_runtime_revision_linkage_cues
    )
)

print(
    "Other cues:",
    len(
        cell109_other_cues
    )
)


print(
    "\n=== TASK-SEMANTIC PARENT CUES ==="
)


for item in (
    cell109_task_parent_semantic_cues
):

    print(
        "-",
        item[
            "root"
        ],
        "::",
        item[
            "path"
        ],
        "=",
        item[
            "value"
        ]
    )


print(
    "\n=== CONCRETE RUNTIME REVISION-LINKAGE FIELDS ==="
)


if (
    cell109_concrete_runtime_field_hits
):

    for item in (
        cell109_concrete_runtime_field_hits
    ):

        print(
            "-",
            item[
                "root"
            ],
            "::",
            item[
                "path"
            ],
            "=",
            item[
                "value"
            ],
            "| patterns:",
            item[
                "matched_patterns"
            ]
        )


else:

    print(
        "No concrete Runtime revision-linkage fields found."
    )


print(
    "\n=== NOTEBOOK HISTORY SEARCH ==="
)

print(
    "History entries inspected:",
    len(
        cell109_input_history
    )
)

print(
    "History term hits:",
    len(
        cell109_history_term_hits
    )
)

print(
    "Construction-like hits:",
    len(
        cell109_history_construction_hits
    )
)


for item in (
    cell109_history_construction_hits
):

    print(
        "\nHistory index:",
        item[
            "history_index"
        ]
    )

    print(
        "Construction lines:"
    )

    for line in (
        item[
            "construction_lines"
        ]
    ):

        print(
            "-",
            line
        )


print(
    "\n=== CORRECTED INTERPRETATION ==="
)

print(
    "Cell108 Parent false positive confirmed:",
    cell109_revision_lineage_interpretation[
        "cell108_parent_false_positive_confirmed"
    ]
)

print(
    "Concrete Runtime revision linkage:",
    cell109_revision_lineage_interpretation[
        "has_concrete_runtime_revision_linkage"
    ]
)

print(
    "Concrete History revision construction:",
    cell109_revision_lineage_interpretation[
        "has_concrete_history_revision_construction"
    ]
)

print(
    "Concrete Revision linkage implementation:",
    cell109_revision_lineage_interpretation[
        "has_concrete_revision_linkage_implementation"
    ]
)

print(
    "Schema status:",
    cell109_revision_lineage_interpretation[
        "schema_status"
    ]
)

print(
    "Next action:",
    cell109_revision_lineage_interpretation[
        "next_action"
    ]
)

print(
    "Inspection SHA256:",
    cell109_revision_lineage_interpretation_sha256
)


print(
    "\n=== FAILURES ==="
)

print(
    "Prerequisite failures:",
    cell109_prerequisite_failures
)

print(
    "History failures:",
    cell109_history_failures
)


print(
    "\nSTATUS:",
    cell109_status
)


print(
    "\n=== CODE CELL 109 GATES ==="
)


for gate_name, gate_value in (
    cell109_gates.items()
):

    print(
        gate_name,
        ":",
        (
            "PASS"
            if gate_value
            else "FAIL"
        )
    )


print(
    "\nFAILED GATES:",
    [
        gate_name

        for gate_name, gate_value
        in cell109_gates.items()

        if not gate_value
    ]
)


print(
    "\nCODE CELL 109 GATE:",
    (
        "PASS"
        if cell_109_pass
        else "FAIL"
    )
)


assert cell_109_pass, (
    "Revision / Lineage refined interpretation failed. "
    "Do not create a Revision or modify the Pilot. "
    "Repair this SAME Code Cell 109 using the printed diagnostics."
)
```

```python
# Code Cell 111

# ============================================================
# Inspect exact Question Draft ID builder implementation
# and the concrete initial Draft Lineage payload.
#
# Read-only only.
# No revision creation.
# No question mutation.
# No evidence mutation.
# No queue mutation.
# No pilot mutation.
# No expected-answer generation.
# No artifact write.
# ============================================================


# ============================================================
# 1) Prerequisites
# ============================================================

cell111_prerequisite_failures = []


if (
    globals().get(
        "cell_110_pass"
    )
    is not True
):

    cell111_prerequisite_failures.append(
        "CELL110_NOT_PASS"
    )


if (
    globals().get(
        "cell110_status"
    )
    !=
    "CELL97_DRAFT_ID_LINEAGE_IMPLEMENTATION_INSPECTED"
):

    cell111_prerequisite_failures.append(
        {
            "reason":
                "CELL110_STATUS_UNEXPECTED",

            "actual":
                globals().get(
                    "cell110_status"
                ),
        }
    )


CELL111_REQUIRED_RUNTIME_OBJECTS = [
    "build_question_draft_id",
    "cell97_question_draft_id",
    "cell97_question_text_sha256",
    "cell97_question_draft_record",
    "cell97_question_draft_lineage",
    "cell97_question_draft_lineage_sha256",
    "cell100_human_adjudication_record",
    "cell95_remaining_pilot_queue",
    "question_drafting_pilot_packets",
    "question_drafting_pilot_by_id",
]


cell111_missing_runtime_objects = [
    name

    for name
    in CELL111_REQUIRED_RUNTIME_OBJECTS

    if (
        name
        not in
        globals()
    )
]


if (
    cell111_missing_runtime_objects
):

    cell111_prerequisite_failures.append(
        {
            "reason":
                "REQUIRED_RUNTIME_OBJECTS_MISSING",

            "objects":
                cell111_missing_runtime_objects,
        }
    )


# ============================================================
# 2) Helpers
# ============================================================

def cell111_sha(value):

    return hashlib.sha256(
        stable_json_key(
            value
        ).encode(
            "utf-8"
        )
    ).hexdigest()


def cell111_line_windows(
    source,
    needles,
    radius=8,
):

    lines = (
        source.splitlines()
    )


    hit_indices = sorted(
        {
            index

            for index, line
            in enumerate(
                lines
            )

            if any(
                needle.lower()
                in
                line.lower()

                for needle
                in needles
            )
        }
    )


    windows = []


    for hit_index in (
        hit_indices
    ):

        start = max(
            0,
            hit_index - radius,
        )

        end = min(
            len(
                lines
            ),
            hit_index + radius + 1,
        )


        windows.append(
            {
                "hit_line_number":
                    hit_index + 1,

                "lines":
                    [
                        {
                            "line_number":
                                line_index + 1,

                            "text":
                                lines[
                                    line_index
                                ],
                        }

                        for line_index
                        in range(
                            start,
                            end,
                        )
                    ],
            }
        )


    return windows


# ============================================================
# 3) Freeze protected state BEFORE inspection
# ============================================================

cell111_question_snapshot_before = (
    cell111_sha(
        cell97_question_draft_record
    )
)


cell111_lineage_snapshot_before = (
    cell111_sha(
        cell97_question_draft_lineage
    )
)


cell111_adjudication_snapshot_before = (
    cell111_sha(
        cell100_human_adjudication_record
    )
)


cell111_queue_snapshot_before = (
    cell111_sha(
        cell95_remaining_pilot_queue
    )
)


cell111_pilot_snapshot_before = (
    cell111_sha(
        {
            "question_drafting_pilot_packets":
                question_drafting_pilot_packets,

            "question_drafting_pilot_by_id":
                question_drafting_pilot_by_id,
        }
    )
)


# ============================================================
# 4) Inspect callable signature
# ============================================================

cell111_signature_failures = []


try:

    import inspect

    cell111_builder_signature = (
        str(
            inspect.signature(
                build_question_draft_id
            )
        )
    )


except Exception as exc:

    cell111_builder_signature = None

    cell111_signature_failures.append(
        {
            "reason":
                "BUILDER_SIGNATURE_INSPECTION_FAILED",

            "error":
                repr(
                    exc
                ),
        }
    )


# ============================================================
# 5) Try direct source inspection
# ============================================================

cell111_direct_source = None
cell111_direct_source_error = None


try:

    cell111_direct_source = (
        inspect.getsource(
            build_question_draft_id
        )
    )


except Exception as exc:

    cell111_direct_source_error = (
        repr(
            exc
        )
    )


# ============================================================
# 6) Read Notebook history for exact builder definition
# ============================================================

cell111_history_failures = []


try:

    cell111_ipython = (
        get_ipython()
    )


except Exception as exc:

    cell111_ipython = None

    cell111_history_failures.append(
        {
            "reason":
                "GET_IPYTHON_FAILED",

            "error":
                repr(
                    exc
                ),
        }
    )


cell111_input_history = []


if (
    cell111_ipython
    is not None
):

    try:

        cell111_input_history = list(
            cell111_ipython.history_manager.input_hist_raw
        )

    except Exception as exc:

        cell111_history_failures.append(
            {
                "reason":
                    "INPUT_HISTORY_READ_FAILED",

                "error":
                    repr(
                        exc
                    ),
            }
        )


cell111_builder_history_candidates = []


for history_index, source in enumerate(
    cell111_input_history
):

    if not isinstance(
        source,
        str,
    ):

        continue


    if (
        "def build_question_draft_id"
        in
        source
    ):

        cell111_builder_history_candidates.append(
            {
                "history_index":
                    history_index,

                "source":
                    source,
            }
        )


cell111_builder_resolution_failures = []


if (
    len(
        cell111_builder_history_candidates
    )
    ==
    0

    and

    cell111_direct_source
    is None
):

    cell111_builder_resolution_failures.append(
        "QUESTION_DRAFT_ID_BUILDER_SOURCE_NOT_FOUND"
    )


# ============================================================
# 7) Resolve source representation conservatively
# ============================================================

if (
    cell111_direct_source
    is not None
):

    cell111_builder_source = (
        cell111_direct_source
    )

    cell111_builder_source_origin = (
        "INSPECT_GETSOURCE"
    )


elif (
    len(
        cell111_builder_history_candidates
    )
    >=
    1
):

    # Prefer the earliest historical definition.
    cell111_builder_history_candidates = sorted(
        cell111_builder_history_candidates,
        key=lambda item:
            item[
                "history_index"
            ],
    )


    cell111_builder_source = (
        cell111_builder_history_candidates[
            0
        ][
            "source"
        ]
    )

    cell111_builder_source_origin = (
        "NOTEBOOK_HISTORY_EARLIEST_DEFINITION"
    )


else:

    cell111_builder_source = ""

    cell111_builder_source_origin = None


# ============================================================
# 8) Extract exact builder implementation window
# ============================================================

cell111_builder_windows = (
    cell111_line_windows(
        cell111_builder_source,
        [
            "def build_question_draft_id",
            "pilot_case_id",
            "question_text_sha256",
            "hashlib",
            "question-draft:",
        ],
        radius=12,
    )
)


# ============================================================
# 9) Inspect current builder behavior without mutation
#
# Rebuild the already-existing ID from the same inputs only.
# ============================================================

cell111_builder_replay_failures = []


try:

    cell111_replayed_existing_id = (
        build_question_draft_id(
            pilot_case_id=
                cell97_question_draft_record[
                    "pilot_case_id"
                ],

            question_text_sha256=
                cell97_question_text_sha256,
        )
    )


except Exception as exc:

    cell111_replayed_existing_id = None

    cell111_builder_replay_failures.append(
        {
            "reason":
                "EXISTING_ID_REPLAY_FAILED",

            "error":
                repr(
                    exc
                ),
        }
    )


cell111_existing_id_is_deterministic = (
    cell111_replayed_existing_id
    ==
    cell97_question_draft_id
)


if not (
    cell111_existing_id_is_deterministic
):

    cell111_builder_replay_failures.append(
        {
            "reason":
                "EXISTING_ID_REPLAY_MISMATCH",

            "expected":
                cell97_question_draft_id,

            "actual":
                cell111_replayed_existing_id,
        }
    )


# ============================================================
# 10) Inspect exact initial Lineage payload
# ============================================================

cell111_lineage_payload = (
    dict(
        cell97_question_draft_lineage
    )
)


cell111_lineage_keys = (
    sorted(
        cell111_lineage_payload.keys()
    )
)


CELL111_IDENTITY_PROVENANCE_FIELDS = [
    "lineage_version",
    "workflow_stage",
    "workflow_contract_sha256",
    "staging_packet_sha256",
    "pilot_case_id",
    "target_evidence_id",
    "target_text_sha256",
    "source_opacity_rule_version",
    "question_draft_id",
    "question_text_sha256",
    "next_action",
]


cell111_missing_expected_lineage_fields = [
    field_name

    for field_name
    in CELL111_IDENTITY_PROVENANCE_FIELDS

    if (
        field_name
        not in
        cell111_lineage_payload
    )
]


cell111_extra_lineage_fields = [
    field_name

    for field_name
    in cell111_lineage_keys

    if (
        field_name
        not in
        CELL111_IDENTITY_PROVENANCE_FIELDS
    )
]


# ============================================================
# 11) Search builder + lineage for revision-linkage primitives
# ============================================================

CELL111_REVISION_LINK_PATTERNS = [
    "revision_id",
    "revision_of",
    "revision_parent",
    "parent_revision",
    "parent_question",
    "previous_question",
    "prior_question",
    "supersedes",
    "superseded_by",
    "replaces_question",
    "replaced_by",
    "lineage_parent",
    "lineage_id",
    "revision_number",
]


cell111_builder_revision_link_hits = [
    pattern

    for pattern
    in CELL111_REVISION_LINK_PATTERNS

    if (
        pattern
        in
        cell111_builder_source.lower()
    )
]


cell111_lineage_revision_link_hits = [
    field_name

    for field_name
    in cell111_lineage_keys

    if any(
        pattern
        in
        field_name.lower()

        for pattern
        in CELL111_REVISION_LINK_PATTERNS
    )
]


# ============================================================
# 12) Determine the builder input contract from signature/source
# ============================================================

cell111_builder_mentions_pilot_case_id = (
    "pilot_case_id"
    in
    cell111_builder_source
)


cell111_builder_mentions_question_text_sha256 = (
    "question_text_sha256"
    in
    cell111_builder_source
)


cell111_builder_mentions_other_revision_input = any(
    pattern
    in
    cell111_builder_source.lower()

    for pattern
    in CELL111_REVISION_LINK_PATTERNS
)


cell111_builder_is_two_identity_input_candidate = (
    cell111_builder_mentions_pilot_case_id
    and
    cell111_builder_mentions_question_text_sha256
    and
    not cell111_builder_mentions_other_revision_input
)


# ============================================================
# 13) Conservative interpretation
# ============================================================

if (
    cell111_existing_id_is_deterministic
    and
    cell111_builder_is_two_identity_input_candidate
    and
    len(
        cell111_builder_revision_link_hits
    )
    ==
    0
    and
    len(
        cell111_lineage_revision_link_hits
    )
    ==
    0
):

    cell111_identity_lineage_interpretation = (
        "EXISTING_PRIMITIVES_SUPPORT_DETERMINISTIC_DRAFT_ID_"
        "BUT_NO_EXPLICIT_REVISION_LINKAGE"
    )

    cell111_next_action = (
        "REASSESS_WHETHER_REVISION_CAN_BE_REPRESENTED_AS_"
        "NEW_DRAFT_ID_PLUS_PRESERVED_EXISTING_LINEAGE_WITHOUT_NEW_FIELDS"
    )


elif (
    len(
        cell111_builder_revision_link_hits
    )
    >
    0

    or

    len(
        cell111_lineage_revision_link_hits
    )
    >
    0
):

    cell111_identity_lineage_interpretation = (
        "EXPLICIT_REVISION_LINKAGE_PRIMITIVE_FOUND"
    )

    cell111_next_action = (
        "INSPECT_EXACT_REVISION_LINKAGE_PRIMITIVE"
    )


else:

    cell111_identity_lineage_interpretation = (
        "DRAFT_ID_OR_LINEAGE_PRIMITIVES_NOT_SUFFICIENTLY_RESOLVED"
    )

    cell111_next_action = (
        "DO_NOT_CREATE_REVISION"
    )


# ============================================================
# 14) Inspection record
# ============================================================

cell111_id_lineage_payload_inspection = {
    "inspection_version":
        "question_draft_id_and_lineage_payload_inspection_v1",

    "builder_signature":
        cell111_builder_signature,

    "builder_source_origin":
        cell111_builder_source_origin,

    "builder_history_candidate_count":
        len(
            cell111_builder_history_candidates
        ),

    "existing_id_replay":
        cell111_replayed_existing_id,

    "existing_id_is_deterministic":
        cell111_existing_id_is_deterministic,

    "builder_mentions_pilot_case_id":
        cell111_builder_mentions_pilot_case_id,

    "builder_mentions_question_text_sha256":
        cell111_builder_mentions_question_text_sha256,

    "builder_mentions_other_revision_input":
        cell111_builder_mentions_other_revision_input,

    "builder_is_two_identity_input_candidate":
        cell111_builder_is_two_identity_input_candidate,

    "lineage_keys":
        list(
            cell111_lineage_keys
        ),

    "missing_expected_lineage_fields":
        list(
            cell111_missing_expected_lineage_fields
        ),

    "extra_lineage_fields":
        list(
            cell111_extra_lineage_fields
        ),

    "builder_revision_link_hits":
        list(
            cell111_builder_revision_link_hits
        ),

    "lineage_revision_link_hits":
        list(
            cell111_lineage_revision_link_hits
        ),

    "interpretation":
        cell111_identity_lineage_interpretation,

    "revision_created":
        False,

    "question_mutated":
        False,

    "evidence_mutated":
        False,

    "queue_mutated":
        False,

    "pilot_mutated":
        False,

    "expected_answer_generated":
        False,

    "next_action":
        cell111_next_action,
}


cell111_id_lineage_payload_inspection_sha256 = (
    cell111_sha(
        cell111_id_lineage_payload_inspection
    )
)


# ============================================================
# 15) Freeze protected state AFTER inspection
# ============================================================

cell111_question_snapshot_after = (
    cell111_sha(
        cell97_question_draft_record
    )
)


cell111_lineage_snapshot_after = (
    cell111_sha(
        cell97_question_draft_lineage
    )
)


cell111_adjudication_snapshot_after = (
    cell111_sha(
        cell100_human_adjudication_record
    )
)


cell111_queue_snapshot_after = (
    cell111_sha(
        cell95_remaining_pilot_queue
    )
)


cell111_pilot_snapshot_after = (
    cell111_sha(
        {
            "question_drafting_pilot_packets":
                question_drafting_pilot_packets,

            "question_drafting_pilot_by_id":
                question_drafting_pilot_by_id,
        }
    )
)


# ============================================================
# 16) Final gates
# ============================================================

cell111_gates = {
    "cell110_is_pass":
        (
            globals().get(
                "cell_110_pass"
            )
            is True
        ),

    "builder_signature_inspected":
        (
            len(
                cell111_signature_failures
            )
            ==
            0
        ),

    "builder_source_resolved":
        (
            bool(
                cell111_builder_source
            )
        ),

    "builder_replay_succeeded":
        (
            len(
                cell111_builder_replay_failures
            )
            ==
            0
        ),

    "existing_draft_id_is_deterministic":
        (
            cell111_existing_id_is_deterministic
        ),

    "lineage_payload_is_complete":
        (
            len(
                cell111_missing_expected_lineage_fields
            )
            ==
            0
        ),

    "lineage_has_no_unexpected_fields":
        (
            len(
                cell111_extra_lineage_fields
            )
            ==
            0
        ),

    "inspection_is_read_only":
        (
            cell111_id_lineage_payload_inspection[
                "revision_created"
            ]
            is False

            and

            cell111_id_lineage_payload_inspection[
                "question_mutated"
            ]
            is False

            and

            cell111_id_lineage_payload_inspection[
                "evidence_mutated"
            ]
            is False

            and

            cell111_id_lineage_payload_inspection[
                "queue_mutated"
            ]
            is False

            and

            cell111_id_lineage_payload_inspection[
                "pilot_mutated"
            ]
            is False
        ),

    "expected_answer_not_generated":
        (
            cell111_id_lineage_payload_inspection[
                "expected_answer_generated"
            ]
            is False
        ),

    "question_record_remains_byte_stable":
        (
            cell111_question_snapshot_before
            ==
            cell111_question_snapshot_after
        ),

    "lineage_record_remains_byte_stable":
        (
            cell111_lineage_snapshot_before
            ==
            cell111_lineage_snapshot_after
        ),

    "adjudication_record_remains_byte_stable":
        (
            cell111_adjudication_snapshot_before
            ==
            cell111_adjudication_snapshot_after
        ),

    "remaining_queue_remains_byte_stable":
        (
            cell111_queue_snapshot_before
            ==
            cell111_queue_snapshot_after
        ),

    "pilot_runtime_remains_byte_stable":
        (
            cell111_pilot_snapshot_before
            ==
            cell111_pilot_snapshot_after
        ),

    "no_artifact_write":
        True,
}


cell_111_pass = all(
    cell111_gates.values()
)


cell111_status = (
    "QUESTION_DRAFT_ID_AND_LINEAGE_PAYLOADS_INSPECTED"

    if (
        cell_111_pass
    )

    else
    "QUESTION_DRAFT_ID_AND_LINEAGE_PAYLOAD_INSPECTION_FAILED"
)


# ============================================================
# 17) Diagnostics
# ============================================================

print(
    "=== QUESTION DRAFT ID / LINEAGE PAYLOAD INSPECTION ==="
)


print(
    "\n=== ID BUILDER ==="
)

print(
    "Signature:",
    cell111_builder_signature
)

print(
    "Source origin:",
    cell111_builder_source_origin
)

print(
    "History definition candidates:",
    len(
        cell111_builder_history_candidates
    )
)

print(
    "Direct getsource error:",
    cell111_direct_source_error
)


print(
    "\n=== BUILDER SOURCE WINDOWS ==="
)


for window_index, window in enumerate(
    cell111_builder_windows,
    start=1,
):

    print(
        f"\n--- Window {window_index} "
        f"(hit line {window['hit_line_number']}) ---"
    )


    for item in (
        window[
            "lines"
        ]
    ):

        print(
            f"{item['line_number']:04d}: "
            f"{item['text']}"
        )


print(
    "\n=== EXISTING ID REPLAY ==="
)

print(
    "Pilot Case ID:",
    cell97_question_draft_record[
        "pilot_case_id"
    ]
)

print(
    "Question text SHA256:",
    cell97_question_text_sha256
)

print(
    "Stored Question Draft ID:",
    cell97_question_draft_id
)

print(
    "Replayed Question Draft ID:",
    cell111_replayed_existing_id
)

print(
    "Deterministic match:",
    cell111_existing_id_is_deterministic
)


print(
    "\n=== LINEAGE PAYLOAD ==="
)

print(
    cell111_lineage_payload
)

print(
    "Lineage keys:",
    cell111_lineage_keys
)

print(
    "Missing expected fields:",
    cell111_missing_expected_lineage_fields
)

print(
    "Extra fields:",
    cell111_extra_lineage_fields
)


print(
    "\n=== REVISION-LINKAGE SEARCH ==="
)

print(
    "Builder revision-link hits:",
    cell111_builder_revision_link_hits
)

print(
    "Lineage revision-link hits:",
    cell111_lineage_revision_link_hits
)

print(
    "Builder mentions pilot_case_id:",
    cell111_builder_mentions_pilot_case_id
)

print(
    "Builder mentions question_text_sha256:",
    cell111_builder_mentions_question_text_sha256
)

print(
    "Builder mentions other revision input:",
    cell111_builder_mentions_other_revision_input
)

print(
    "Builder is two-identity-input candidate:",
    cell111_builder_is_two_identity_input_candidate
)


print(
    "\n=== INTERPRETATION ==="
)

print(
    "Interpretation:",
    cell111_identity_lineage_interpretation
)

print(
    "Next action:",
    cell111_next_action
)

print(
    "Inspection SHA256:",
    cell111_id_lineage_payload_inspection_sha256
)


print(
    "\n=== FAILURES ==="
)

print(
    "Prerequisite failures:",
    cell111_prerequisite_failures
)

print(
    "Signature failures:",
    cell111_signature_failures
)

print(
    "History failures:",
    cell111_history_failures
)

print(
    "Builder resolution failures:",
    cell111_builder_resolution_failures
)

print(
    "Builder replay failures:",
    cell111_builder_replay_failures
)


print(
    "\nSTATUS:",
    cell111_status
)


print(
    "\n=== CODE CELL 111 GATES ==="
)


for gate_name, gate_value in (
    cell111_gates.items()
):

    print(
        gate_name,
        ":",
        (
            "PASS"
            if gate_value
            else "FAIL"
        )
    )


print(
    "\nFAILED GATES:",
    [
        gate_name

        for gate_name, gate_value
        in cell111_gates.items()

        if not gate_value
    ]
)


print(
    "\nCODE CELL 111 GATE:",
    (
        "PASS"
        if cell_111_pass
        else "FAIL"
    )
)


assert cell_111_pass, (
    "Question Draft ID / Lineage payload inspection failed. "
    "Do not create a Revision or modify the Pilot. "
    "Repair this SAME Code Cell 111 using the printed diagnostics."
)
```

<h3>بررسی وجود Registry / History واقعی برای حفظ Question Revision Lineage</h3>

<p>
Cell 111 با موفقیت PASS شد و نشان داد
<code>build_question_draft_id</code>
فقط از
<code>pilot_case_id</code>
و
<code>question_text_sha256</code>
استفاده می‌کند؛ بنابراین تغییر متن Question می‌تواند به‌صورت Deterministic یک Draft ID جدید ایجاد کند.
در عین حال، نه ID Builder و نه Lineage فعلی هیچ Field صریحی برای اتصال Revision جدید به Draft قبلی ندارند. :contentReference[oaicite:0]{index=0}
</p>

<p>
قبل از تصمیم درباره ساخت Revision، باید مشخص شود آیا در Notebook یا Runtime فعلی یک Registry، History Collection یا Pattern واقعی برای نگهداری چند Question Draft / Lineage و حفظ نسخه‌های قبلی وجود دارد یا خیر.

<p>
این مرحله فقط Read-Only بررسی می‌کند:
</p>

<ul>
<li>Globalهایی که Question Draft یا Lineageهای متعدد را نگه می‌دارند؛</li>
<li>Collectionهایی مانند List / Dict / Registry / History برای Draftها؛</li>
<li>Patternهای <code>append</code>، <code>register</code>، <code>history</code> یا ذخیره چند Revision در Notebook History؛</li>
<li>اینکه آیا «Preserve Lineage» را می‌توان با Primitiveهای موجود و بدون افزودن Field جدید اجرا کرد.</li>
</ul>

<p>
وجود صرف Draft ID جدید کافی نیست.
برای اجرای Rule
<code>CREATE_NEW_REVISION_ID_PRESERVE_LINEAGE</code>
باید یک مکانیزم واقعی برای نگهداری Draft قبلی و ارتباط قابل اثبات بین نسخه‌ها وجود داشته باشد.
</p>

<p>
در این Cell هیچ Revision ساخته نمی‌شود، Question و Evidence تغییر نمی‌کنند، Queue جلو نمی‌رود و Expected Answer همچنان Blocked می‌ماند.
</p>

<h4>معیار PASS</h4>

<p>
Runtime و Notebook History باید بدون Mutation برای Registry/History واقعی Revision بررسی شوند.
اگر هیچ مکانیزم قابل استفاده‌ای پیدا نشود، نتیجه باید صریحاً
<code>NO_EXISTING_REVISION_HISTORY_MECHANISM_FOUND</code>
باشد و هیچ Schema یا Registry جدیدی ساخته نشود.
</p>

```python
# Code Cell 112

# ============================================================
# Inspect existing Runtime / Notebook history for a real
# Question Draft revision history or lineage registry.
#
# Goal:
# Determine whether CREATE_NEW_REVISION_ID_PRESERVE_LINEAGE
# can be implemented using existing primitives only.
#
# Read-only only.
# No revision creation.
# No registry creation.
# No question mutation.
# No evidence mutation.
# No queue mutation.
# No pilot mutation.
# No expected-answer generation.
# No artifact write.
# ============================================================


# ============================================================
# 1) Prerequisites
# ============================================================

cell112_prerequisite_failures = []


if (
    globals().get(
        "cell_111_pass"
    )
    is not True
):

    cell112_prerequisite_failures.append(
        "CELL111_NOT_PASS"
    )


if (
    globals().get(
        "cell111_status"
    )
    !=
    "QUESTION_DRAFT_ID_AND_LINEAGE_PAYLOADS_INSPECTED"
):

    cell112_prerequisite_failures.append(
        {
            "reason":
                "CELL111_STATUS_UNEXPECTED",

            "actual":
                globals().get(
                    "cell111_status"
                ),
        }
    )


CELL112_REQUIRED_RUNTIME_OBJECTS = [
    "cell97_question_draft_record",
    "cell97_question_draft_lineage",
    "cell97_question_draft_id",
    "cell100_human_adjudication_record",
    "cell95_remaining_pilot_queue",
    "question_drafting_pilot_packets",
    "question_drafting_pilot_by_id",
    "PILOT_QA_WORKFLOW_CONTRACT_V1",
]


cell112_missing_runtime_objects = [
    name

    for name
    in CELL112_REQUIRED_RUNTIME_OBJECTS

    if (
        name
        not in
        globals()
    )
]


if (
    cell112_missing_runtime_objects
):

    cell112_prerequisite_failures.append(
        {
            "reason":
                "REQUIRED_RUNTIME_OBJECTS_MISSING",

            "objects":
                cell112_missing_runtime_objects,
        }
    )


# ============================================================
# 2) Helpers
# ============================================================

def cell112_sha(value):

    return hashlib.sha256(
        stable_json_key(
            value
        ).encode(
            "utf-8"
        )
    ).hexdigest()


def cell112_walk(value, path=()):

    if isinstance(
        value,
        dict,
    ):

        for key, child in value.items():

            yield from cell112_walk(
                child,
                path + (str(key),),
            )

    elif isinstance(
        value,
        (list, tuple),
    ):

        for index, child in enumerate(
            value
        ):

            yield from cell112_walk(
                child,
                path + (str(index),),
            )

    else:

        yield path, value


# ============================================================
# 3) Freeze protected state BEFORE inspection
# ============================================================

cell112_question_snapshot_before = (
    cell112_sha(
        cell97_question_draft_record
    )
)


cell112_lineage_snapshot_before = (
    cell112_sha(
        cell97_question_draft_lineage
    )
)


cell112_adjudication_snapshot_before = (
    cell112_sha(
        cell100_human_adjudication_record
    )
)


cell112_queue_snapshot_before = (
    cell112_sha(
        cell95_remaining_pilot_queue
    )
)


cell112_pilot_snapshot_before = (
    cell112_sha(
        {
            "question_drafting_pilot_packets":
                question_drafting_pilot_packets,

            "question_drafting_pilot_by_id":
                question_drafting_pilot_by_id,
        }
    )
)


cell112_contract_snapshot_before = (
    cell112_sha(
        PILOT_QA_WORKFLOW_CONTRACT_V1
    )
)


# ============================================================
# 4) Search Globals for revision/history/registry candidates
# ============================================================

CELL112_GLOBAL_NAME_TERMS = [
    "question",
    "draft",
    "lineage",
    "revision",
    "history",
    "registry",
    "versions",
    "records",
]


cell112_global_collection_candidates = []


for name, value in list(
    globals().items()
):

    name_text = (
        str(
            name
        )
    )


    name_lower = (
        name_text.lower()
    )


    if (
        name_text.startswith(
            "__"
        )
    ):

        continue


    if not any(
        term
        in
        name_lower

        for term
        in CELL112_GLOBAL_NAME_TERMS
    ):

        continue


    if not isinstance(
        value,
        (
            list,
            tuple,
            dict,
        ),
    ):

        continue


    cell112_global_collection_candidates.append(
        {
            "name":
                name_text,

            "type":
                type(
                    value
                ).__name__,

            "size":
                len(
                    value
                ),

            "value":
                value,
        }
    )


# ============================================================
# 5) Inspect candidate collections for Question Draft IDs
# ============================================================

cell112_question_draft_collection_hits = []


for candidate in (
    cell112_global_collection_candidates
):

    draft_ids = sorted(
        {
            scalar

            for _, scalar
            in cell112_walk(
                candidate[
                    "value"
                ]
            )

            if (
                isinstance(
                    scalar,
                    str,
                )

                and

                scalar.startswith(
                    "question-draft:"
                )
            )
        }
    )


    lineage_versions = sorted(
        {
            scalar

            for _, scalar
            in cell112_walk(
                candidate[
                    "value"
                ]
            )

            if (
                isinstance(
                    scalar,
                    str,
                )

                and

                "question_draft_lineage"
                in
                scalar
            )
        }
    )


    if (
        draft_ids
        or
        lineage_versions
    ):

        cell112_question_draft_collection_hits.append(
            {
                "name":
                    candidate[
                        "name"
                    ],

                "type":
                    candidate[
                        "type"
                    ],

                "size":
                    candidate[
                        "size"
                    ],

                "question_draft_ids":
                    draft_ids,

                "lineage_versions":
                    lineage_versions,
            }
        )


# ============================================================
# 6) Identify collections that contain multiple Draft IDs
# ============================================================

cell112_multi_draft_collections = [
    item

    for item
    in cell112_question_draft_collection_hits

    if (
        len(
            item[
                "question_draft_ids"
            ]
        )
        >
        1
    )
]


cell112_current_draft_registered_collections = [
    item

    for item
    in cell112_question_draft_collection_hits

    if (
        cell97_question_draft_id
        in
        item[
            "question_draft_ids"
        ]
    )
]


# ============================================================
# 7) Search for exact Lineage object registration
# ============================================================

cell112_lineage_sha = (
    cell112_sha(
        cell97_question_draft_lineage
    )
)


cell112_exact_lineage_registry_hits = []


for candidate in (
    cell112_global_collection_candidates
):

    value = (
        candidate[
            "value"
        ]
    )


    exact_object_found = False


    if isinstance(
        value,
        dict,
    ):

        exact_object_found = any(
            child
            ==
            cell97_question_draft_lineage

            for child
            in value.values()
        )


    elif isinstance(
        value,
        (
            list,
            tuple,
        ),
    ):

        exact_object_found = any(
            child
            ==
            cell97_question_draft_lineage

            for child
            in value
        )


    if (
        exact_object_found
    ):

        cell112_exact_lineage_registry_hits.append(
            {
                "name":
                    candidate[
                        "name"
                    ],

                "type":
                    candidate[
                        "type"
                    ],

                "size":
                    candidate[
                        "size"
                    ],
            }
        )


# ============================================================
# 8) Read Notebook history
# ============================================================

cell112_history_failures = []


try:

    cell112_ipython = (
        get_ipython()
    )


except Exception as exc:

    cell112_ipython = None

    cell112_history_failures.append(
        {
            "reason":
                "GET_IPYTHON_FAILED",

            "error":
                repr(
                    exc
                ),
        }
    )


cell112_input_history = []


if (
    cell112_ipython
    is not None
):

    try:

        cell112_input_history = list(
            cell112_ipython.history_manager.input_hist_raw
        )

    except Exception as exc:

        cell112_history_failures.append(
            {
                "reason":
                    "INPUT_HISTORY_READ_FAILED",

                "error":
                    repr(
                        exc
                    ),
            }
        )


# ============================================================
# 9) Search history for real registration / append patterns
# ============================================================

CELL112_HISTORY_OBJECT_TERMS = [
    "question_draft",
    "draft_lineage",
    "question-draft:",
]


CELL112_HISTORY_ACTION_TERMS = [
    ".append(",
    ".extend(",
    ".update(",
    "setdefault(",
    "register",
    "registry",
    "history",
    "versions",
    "revisions",
]


cell112_history_registration_hits = []


for history_index, source in enumerate(
    cell112_input_history
):

    if not isinstance(
        source,
        str,
    ):

        continue


    source_lower = (
        source.lower()
    )


    object_hits = [
        term

        for term
        in CELL112_HISTORY_OBJECT_TERMS

        if (
            term
            in
            source_lower
        )
    ]


    action_hits = [
        term

        for term
        in CELL112_HISTORY_ACTION_TERMS

        if (
            term
            in
            source_lower
        )
    ]


    if not (
        object_hits
        and
        action_hits
    ):

        continue


    relevant_lines = [
        line.strip()

        for line
        in source.splitlines()

        if (
            any(
                term
                in
                line.lower()

                for term
                in object_hits
            )

            or

            any(
                term
                in
                line.lower()

                for term
                in action_hits
            )
        )
    ]


    cell112_history_registration_hits.append(
        {
            "history_index":
                history_index,

            "object_hits":
                object_hits,

            "action_hits":
                action_hits,

            "relevant_lines":
                relevant_lines[
                    :30
                ],
        }
    )


# ============================================================
# 10) Remove analysis-cell self references conservatively
# ============================================================

cell112_history_registration_hits_filtered = []


for item in (
    cell112_history_registration_hits
):

    source = (
        cell112_input_history[
            item[
                "history_index"
            ]
        ]
    )


    is_current_or_later_analysis_cell = any(
        marker
        in
        source

        for marker
        in [
            "# Code Cell 108",
            "# Code Cell 109",
            "# Code Cell 110",
            "# Code Cell 111",
            "# Code Cell 112",
        ]
    )


    if not (
        is_current_or_later_analysis_cell
    ):

        cell112_history_registration_hits_filtered.append(
            item
        )


# ============================================================
# 11) Search for explicit Revision History concepts in contract
# ============================================================

cell112_contract_revision_history_hits = []


for path, scalar in cell112_walk(
    PILOT_QA_WORKFLOW_CONTRACT_V1
):

    path_text = (
        ".".join(
            path
        )
    )


    scalar_text = (
        str(
            scalar
        )
    )


    combined = (
        f"{path_text}\n{scalar_text}"
    ).lower()


    matched_terms = [
        term

        for term
        in [
            "revision",
            "lineage",
            "history",
            "registry",
            "versions",
            "revisions",
        ]

        if (
            term
            in
            combined
        )
    ]


    if (
        matched_terms
    ):

        cell112_contract_revision_history_hits.append(
            {
                "path":
                    path_text,

                "value":
                    scalar,

                "matched_terms":
                    matched_terms,
            }
        )


# ============================================================
# 12) Determine whether an existing Revision History
#     mechanism is actually available
# ============================================================

cell112_has_multi_draft_runtime_collection = (
    len(
        cell112_multi_draft_collections
    )
    >
    0
)


cell112_has_exact_lineage_registry = (
    len(
        cell112_exact_lineage_registry_hits
    )
    >
    0
)


cell112_has_prior_history_registration_pattern = (
    len(
        cell112_history_registration_hits_filtered
    )
    >
    0
)


cell112_has_existing_revision_history_mechanism = (
    cell112_has_multi_draft_runtime_collection
    or
    cell112_has_exact_lineage_registry
    or
    cell112_has_prior_history_registration_pattern
)


# ============================================================
# 13) Conservative interpretation
# ============================================================

if (
    cell112_has_existing_revision_history_mechanism
):

    cell112_mechanism_status = (
        "EXISTING_REVISION_HISTORY_MECHANISM_CANDIDATE_FOUND"
    )

    cell112_next_action = (
        "INSPECT_EXACT_EXISTING_HISTORY_MECHANISM_BEFORE_CREATING_REVISION"
    )


else:

    cell112_mechanism_status = (
        "NO_EXISTING_REVISION_HISTORY_MECHANISM_FOUND"
    )

    cell112_next_action = (
        "DO_NOT_CREATE_REVISION_OR_NEW_REGISTRY_"
        "CURRENT_PILOT_REMAINS_BLOCKED_PENDING_EXPLICIT_WORKFLOW_DECISION"
    )


# ============================================================
# 14) Inspection record
# ============================================================

cell112_revision_history_inspection = {
    "inspection_version":
        "question_revision_history_mechanism_inspection_v1",

    "pilot_case_id":
        cell100_human_adjudication_record[
            "pilot_case_id"
        ],

    "current_question_draft_id":
        cell97_question_draft_id,

    "revision_policy":
        PILOT_QA_WORKFLOW_CONTRACT_V1[
            "question_revision_policy"
        ],

    "global_collection_candidate_count":
        len(
            cell112_global_collection_candidates
        ),

    "question_draft_collection_hit_count":
        len(
            cell112_question_draft_collection_hits
        ),

    "multi_draft_collection_count":
        len(
            cell112_multi_draft_collections
        ),

    "current_draft_registered_collection_count":
        len(
            cell112_current_draft_registered_collections
        ),

    "exact_lineage_registry_count":
        len(
            cell112_exact_lineage_registry_hits
        ),

    "history_registration_hit_count":
        len(
            cell112_history_registration_hits_filtered
        ),

    "contract_revision_history_cue_count":
        len(
            cell112_contract_revision_history_hits
        ),

    "has_multi_draft_runtime_collection":
        cell112_has_multi_draft_runtime_collection,

    "has_exact_lineage_registry":
        cell112_has_exact_lineage_registry,

    "has_prior_history_registration_pattern":
        cell112_has_prior_history_registration_pattern,

    "has_existing_revision_history_mechanism":
        cell112_has_existing_revision_history_mechanism,

    "mechanism_status":
        cell112_mechanism_status,

    "revision_created":
        False,

    "registry_created":
        False,

    "question_mutated":
        False,

    "evidence_mutated":
        False,

    "queue_mutated":
        False,

    "pilot_mutated":
        False,

    "expected_answer_generated":
        False,

    "next_action":
        cell112_next_action,
}


cell112_revision_history_inspection_sha256 = (
    cell112_sha(
        cell112_revision_history_inspection
    )
)


# ============================================================
# 15) Freeze protected state AFTER inspection
# ============================================================

cell112_question_snapshot_after = (
    cell112_sha(
        cell97_question_draft_record
    )
)


cell112_lineage_snapshot_after = (
    cell112_sha(
        cell97_question_draft_lineage
    )
)


cell112_adjudication_snapshot_after = (
    cell112_sha(
        cell100_human_adjudication_record
    )
)


cell112_queue_snapshot_after = (
    cell112_sha(
        cell95_remaining_pilot_queue
    )
)


cell112_pilot_snapshot_after = (
    cell112_sha(
        {
            "question_drafting_pilot_packets":
                question_drafting_pilot_packets,

            "question_drafting_pilot_by_id":
                question_drafting_pilot_by_id,
        }
    )
)


cell112_contract_snapshot_after = (
    cell112_sha(
        PILOT_QA_WORKFLOW_CONTRACT_V1
    )
)


# ============================================================
# 16) Final gates
# ============================================================

cell112_gates = {
    "cell111_is_pass":
        (
            globals().get(
                "cell_111_pass"
            )
            is True
        ),

    "revision_policy_is_exact":
        (
            PILOT_QA_WORKFLOW_CONTRACT_V1.get(
                "question_revision_policy"
            )
            ==
            "CREATE_NEW_REVISION_ID_PRESERVE_LINEAGE"
        ),

    "runtime_collections_were_inspected":
        (
            isinstance(
                cell112_global_collection_candidates,
                list,
            )
        ),

    "notebook_history_read_succeeded":
        (
            len(
                cell112_history_failures
            )
            ==
            0
        ),

    "history_registration_patterns_were_inspected":
        (
            isinstance(
                cell112_history_registration_hits_filtered,
                list,
            )
        ),

    "contract_revision_history_cues_were_inspected":
        (
            isinstance(
                cell112_contract_revision_history_hits,
                list,
            )
        ),

    "no_revision_created":
        (
            cell112_revision_history_inspection[
                "revision_created"
            ]
            is False
        ),

    "no_registry_created":
        (
            cell112_revision_history_inspection[
                "registry_created"
            ]
            is False
        ),

    "question_not_mutated":
        (
            cell112_revision_history_inspection[
                "question_mutated"
            ]
            is False
        ),

    "evidence_not_mutated":
        (
            cell112_revision_history_inspection[
                "evidence_mutated"
            ]
            is False
        ),

    "queue_not_mutated":
        (
            cell112_revision_history_inspection[
                "queue_mutated"
            ]
            is False
        ),

    "pilot_not_mutated":
        (
            cell112_revision_history_inspection[
                "pilot_mutated"
            ]
            is False
        ),

    "expected_answer_not_generated":
        (
            cell112_revision_history_inspection[
                "expected_answer_generated"
            ]
            is False
        ),

    "question_record_remains_byte_stable":
        (
            cell112_question_snapshot_before
            ==
            cell112_question_snapshot_after
        ),

    "lineage_record_remains_byte_stable":
        (
            cell112_lineage_snapshot_before
            ==
            cell112_lineage_snapshot_after
        ),

    "adjudication_record_remains_byte_stable":
        (
            cell112_adjudication_snapshot_before
            ==
            cell112_adjudication_snapshot_after
        ),

    "remaining_queue_remains_byte_stable":
        (
            cell112_queue_snapshot_before
            ==
            cell112_queue_snapshot_after
        ),

    "pilot_runtime_remains_byte_stable":
        (
            cell112_pilot_snapshot_before
            ==
            cell112_pilot_snapshot_after
        ),

    "workflow_contract_remains_byte_stable":
        (
            cell112_contract_snapshot_before
            ==
            cell112_contract_snapshot_after
        ),

    "no_artifact_write":
        True,
}


cell_112_pass = all(
    cell112_gates.values()
)


cell112_status = (
    "REVISION_HISTORY_MECHANISM_INSPECTED"

    if (
        cell_112_pass
    )

    else
    "REVISION_HISTORY_MECHANISM_INSPECTION_FAILED"
)


# ============================================================
# 17) Diagnostics
# ============================================================

print(
    "=== QUESTION REVISION HISTORY / REGISTRY INSPECTION ==="
)


print(
    "\n=== CURRENT STATE ==="
)

print(
    "Pilot ID:",
    cell100_human_adjudication_record[
        "pilot_case_id"
    ]
)

print(
    "Current Question Draft ID:",
    cell97_question_draft_id
)

print(
    "Revision policy:",
    PILOT_QA_WORKFLOW_CONTRACT_V1[
        "question_revision_policy"
    ]
)


print(
    "\n=== GLOBAL COLLECTION DISCOVERY ==="
)

print(
    "Collection candidates:",
    len(
        cell112_global_collection_candidates
    )
)

print(
    "Collections containing Question Draft IDs / Lineage:",
    len(
        cell112_question_draft_collection_hits
    )
)

print(
    "Multi-Draft collections:",
    len(
        cell112_multi_draft_collections
    )
)

print(
    "Collections containing current Draft:",
    len(
        cell112_current_draft_registered_collections
    )
)

print(
    "Exact Lineage registries:",
    len(
        cell112_exact_lineage_registry_hits
    )
)


print(
    "\n=== QUESTION DRAFT COLLECTION HITS ==="
)


if (
    cell112_question_draft_collection_hits
):

    for item in (
        cell112_question_draft_collection_hits
    ):

        print(
            "\nCollection:",
            item[
                "name"
            ]
        )

        print(
            "Type:",
            item[
                "type"
            ]
        )

        print(
            "Size:",
            item[
                "size"
            ]
        )

        print(
            "Question Draft IDs:",
            item[
                "question_draft_ids"
            ]
        )

        print(
            "Lineage versions:",
            item[
                "lineage_versions"
            ]
        )


else:

    print(
        "No Runtime collection containing Question Draft IDs or Lineage was found."
    )


print(
    "\n=== EXACT LINEAGE REGISTRY HITS ==="
)


if (
    cell112_exact_lineage_registry_hits
):

    for item in (
        cell112_exact_lineage_registry_hits
    ):

        print(
            item
        )


else:

    print(
        "No collection registers the current Lineage object."
    )


print(
    "\n=== NOTEBOOK HISTORY REGISTRATION PATTERNS ==="
)

print(
    "Raw matching history entries:",
    len(
        cell112_history_registration_hits
    )
)

print(
    "Filtered prior implementation hits:",
    len(
        cell112_history_registration_hits_filtered
    )
)


if (
    cell112_history_registration_hits_filtered
):

    for item in (
        cell112_history_registration_hits_filtered
    ):

        print(
            "\nHistory index:",
            item[
                "history_index"
            ]
        )

        print(
            "Object hits:",
            item[
                "object_hits"
            ]
        )

        print(
            "Action hits:",
            item[
                "action_hits"
            ]
        )

        print(
            "Relevant lines:"
        )


        for line in (
            item[
                "relevant_lines"
            ]
        ):

            print(
                "-",
                line
            )


else:

    print(
        "No prior Question Draft / Lineage registration implementation found."
    )


print(
    "\n=== CONTRACT REVISION / HISTORY CUES ==="
)


for item in (
    cell112_contract_revision_history_hits
):

    print(
        "-",
        item[
            "path"
        ],
        "=",
        item[
            "value"
        ],
        "|",
        item[
            "matched_terms"
        ]
    )


print(
    "\n=== INTERPRETATION ==="
)

print(
    "Has multi-Draft Runtime collection:",
    cell112_has_multi_draft_runtime_collection
)

print(
    "Has exact Lineage registry:",
    cell112_has_exact_lineage_registry
)

print(
    "Has prior History registration pattern:",
    cell112_has_prior_history_registration_pattern
)

print(
    "Has existing Revision History mechanism:",
    cell112_has_existing_revision_history_mechanism
)

print(
    "Mechanism status:",
    cell112_mechanism_status
)

print(
    "Next action:",
    cell112_next_action
)

print(
    "Inspection SHA256:",
    cell112_revision_history_inspection_sha256
)


print(
    "\n=== FAILURES ==="
)

print(
    "Prerequisite failures:",
    cell112_prerequisite_failures
)

print(
    "History failures:",
    cell112_history_failures
)


print(
    "\nSTATUS:",
    cell112_status
)


print(
    "\n=== CODE CELL 112 GATES ==="
)


for gate_name, gate_value in (
    cell112_gates.items()
):

    print(
        gate_name,
        ":",
        (
            "PASS"
            if gate_value
            else "FAIL"
        )
    )


print(
    "\nFAILED GATES:",
    [
        gate_name

        for gate_name, gate_value
        in cell112_gates.items()

        if not gate_value
    ]
)


print(
    "\nCODE CELL 112 GATE:",
    (
        "PASS"
        if cell_112_pass
        else "FAIL"
    )
)


assert cell_112_pass, (
    "Revision History / Registry inspection failed. "
    "Do not create a Revision, Registry, or modify the Pilot. "
    "Repair this SAME Code Cell 112 using the printed diagnostics."
)
```

<h3>بررسی دقیق Revision Mechanism واقعی موجود در Pilot مرجع</h3>

<p>
Cell 112 با Gate کامل PASS شد و نشان داد یک Candidate واقعی و مهم از Revision قبلی در Runtime وجود دارد:
<code>cell86_revised_question_draft_record</code>

<code>cell86_revision_lineage</code>.
در همان Runtime، Lineage مربوط به Cell 86 شامل هر دو Question Draft ID قبلی و Revised است؛ بنابراین این مورد باید به‌عنوان Reference Implementation واقعی بررسی شود، نه Patternهای عمومی <code>append</code> که می‌توانند False Positive باشند. :contentReference[oaicite:0]{index=0}
</p>

<p>
این مرحله فقط Reference Revision قبلی را Read-Only بررسی می‌کند تا دقیقاً مشخص شود:
</p>

<ul>
<li>Revision در Cell 86 چگونه ساخته شده است؛</li>
<li>Draft قبلی و Draft جدید با چه Fieldهایی به هم متصل شده‌اند؛</li>

```python
# Code Cell 118

# ============================================================
# SAME-CELL REPAIR
#
# Distinguish harmless LOCAL mutation such as:
#     candidates.append(...)
#
# from EXTERNAL / Runtime state mutation.
#
# Read-only only.
# Do not call select_website_local_target.
# Do not call select_target_evidence.
# Do not build or register Target Evidence.
# No artifact write.
# ============================================================

import inspect
import ast
import textwrap


# ============================================================
# 1) Prerequisites
# ============================================================

cell118_prerequisite_failures = []


if globals().get(
    "cell_117_pass"
) is not True:

    cell118_prerequisite_failures.append(
        "CELL117_NOT_PASS"
    )


if (
    globals().get(
        "cell117_status"
    )
    !=
    "CURRENT_TARGET_SELECTION_BRANCH_CONTRACT_INSPECTED"
):

    cell118_prerequisite_failures.append({
        "reason":
            "CELL117_STATUS_UNEXPECTED",

        "actual":
            globals().get(
                "cell117_status"
            ),
    })


CELL118_REQUIRED_OBJECTS = [
    "select_website_local_target",
    "cell117_candidate",
    "cell117_method",
    "cell97_question_draft_record",
    "target_evidence_records",
    "target_evidence_by_case_id",
    "target_evidence_by_task_draft_id",
    "cell95_remaining_pilot_queue",
    "question_drafting_pilot_packets",
    "question_drafting_pilot_by_id",
]


cell118_missing_objects = [
    name

    for name
    in CELL118_REQUIRED_OBJECTS

    if name not in globals()
]


if cell118_missing_objects:

    cell118_prerequisite_failures.append({
        "reason":
            "REQUIRED_RUNTIME_OBJECTS_MISSING",

        "objects":
            cell118_missing_objects,
    })


# ============================================================
# 2) Helper
# ============================================================

def cell118_sha(value):

    return hashlib.sha256(
        stable_json_key(
            value
        ).encode(
            "utf-8"
        )
    ).hexdigest()


# ============================================================
# 3) Freeze protected state BEFORE inspection
# ============================================================

cell118_before = {
    "question":
        cell118_sha(
            cell97_question_draft_record
        ),

    "targets":
        cell118_sha(
            target_evidence_records
        ),

    "case_index":
        cell118_sha(
            target_evidence_by_case_id
        ),

    "task_index":
        cell118_sha(
            target_evidence_by_task_draft_id
        ),

    "queue":
        cell118_sha(
            cell95_remaining_pilot_queue
        ),

    "pilot":
        cell118_sha({
            "packets":
                question_drafting_pilot_packets,

            "by_id":
                question_drafting_pilot_by_id,
        }),
}


# ============================================================
# 4) Resolve helper signature and source
# ============================================================

cell118_inspection_failures = []


try:

    cell118_helper_signature = str(
        inspect.signature(
            select_website_local_target
        )
    )


except Exception as exc:

    cell118_helper_signature = None

    cell118_inspection_failures.append({
        "reason":
            "HELPER_SIGNATURE_FAILED",

        "error":
            repr(
                exc
            ),
    })


try:

    cell118_helper_source = (
        inspect.getsource(
            select_website_local_target
        )
    )


except Exception as exc:

    cell118_helper_source = None

    cell118_inspection_failures.append({
        "reason":
            "HELPER_SOURCE_FAILED",

        "error":
            repr(
                exc
            ),
    })


# ============================================================
# 5) Parse helper AST
# ============================================================

cell118_helper_tree = None
cell118_function_node = None


if isinstance(
    cell118_helper_source,
    str,
):

    try:

        cell118_helper_tree = ast.parse(
            textwrap.dedent(
                cell118_helper_source
            )
        )


        function_nodes = [
            node

            for node
            in cell118_helper_tree.body

            if isinstance(
                node,
                (
                    ast.FunctionDef,
                    ast.AsyncFunctionDef,
                ),
            )
        ]


        if (
            len(
                function_nodes
            )
            ==
            1
        ):

            cell118_function_node = (
                function_nodes[
                    0
                ]
            )


        else:

            cell118_inspection_failures.append({
                "reason":
                    "HELPER_FUNCTION_NODE_NOT_UNIQUE",

                "count":
                    len(
                        function_nodes
                    ),
            })


    except Exception as exc:

        cell118_inspection_failures.append({
            "reason":
                "HELPER_AST_PARSE_FAILED",

            "error":
                repr(
                    exc
                ),
        })


# ============================================================
# 6) Resolve local variable names
# ============================================================

cell118_local_names = set()


if (
    cell118_function_node
    is not None
):

    # Function parameters are local.
    for arg in (
        list(
            cell118_function_node.args.args
        )
        +
        list(
            cell118_function_node.args.kwonlyargs
        )
    ):

        cell118_local_names.add(
            arg.arg
        )


    if (
        cell118_function_node.args.vararg
        is not None
    ):

        cell118_local_names.add(
            cell118_function_node.args.vararg.arg
        )


    if (
        cell118_function_node.args.kwarg
        is not None
    ):

        cell118_local_names.add(
            cell118_function_node.args.kwarg.arg
        )


    # Assigned names are local unless declared global/nonlocal.
    for node in ast.walk(
        cell118_function_node
    ):

        if isinstance(
            node,
            ast.Name,
        ) and isinstance(
            node.ctx,
            ast.Store,
        ):

            cell118_local_names.add(
                node.id
            )


cell118_global_declarations = set()
cell118_nonlocal_declarations = set()


if (
    cell118_function_node
    is not None
):

    for node in ast.walk(
        cell118_function_node
    ):

        if isinstance(
            node,
            ast.Global,
        ):

            cell118_global_declarations.update(
                node.names
            )


        elif isinstance(
            node,
            ast.Nonlocal,
        ):

            cell118_nonlocal_declarations.update(
                node.names
            )


cell118_local_names = (
    cell118_local_names
    -
    cell118_global_declarations
    -
    cell118_nonlocal_declarations
)


# ============================================================
# 7) Detect mutation calls and classify receiver scope
# ============================================================

CELL118_MUTATING_METHODS = {
    "append",
    "extend",
    "insert",
    "remove",
    "pop",
    "clear",
    "sort",
    "reverse",
    "update",
    "setdefault",
    "add",
    "discard",
}


cell118_local_mutation_calls = []
cell118_external_mutation_calls = []


if (
    cell118_function_node
    is not None
):

    for node in ast.walk(
        cell118_function_node
    ):

        if not isinstance(
            node,
            ast.Call,
        ):

            continue


        if not isinstance(
            node.func,
            ast.Attribute,
        ):

            continue


        method_name = (
            node.func.attr
        )


        if (
            method_name
            not in
            CELL118_MUTATING_METHODS
        ):

            continue


        receiver = (
            node.func.value
        )


        receiver_name = (
            receiver.id

            if isinstance(
                receiver,
                ast.Name,
            )

            else None
        )


        record = {
            "method":
                method_name,

            "receiver":
                receiver_name,

            "line":
                getattr(
                    node,
                    "lineno",
                    None,
                ),
        }


        if (
            receiver_name
            in
            cell118_local_names
        ):

            cell118_local_mutation_calls.append(
                record
            )


        else:

            cell118_external_mutation_calls.append(
                record
            )


# ============================================================
# 8) Detect direct assignment to external objects
# ============================================================

cell118_external_subscript_assignments = []


if (
    cell118_function_node
    is not None
):

    for node in ast.walk(
        cell118_function_node
    ):

        if not isinstance(
            node,
            (
                ast.Assign,
                ast.AnnAssign,
                ast.AugAssign,
            ),
        ):

            continue


        targets = []


        if isinstance(
            node,
            ast.Assign,
        ):

            targets = list(
                node.targets
            )


        else:

            targets = [
                node.target
            ]


        for target in targets:

            if not isinstance(
                target,
                ast.Subscript,
            ):

                continue


            root = (
                target.value
            )


            root_name = (
                root.id

                if isinstance(
                    root,
                    ast.Name,
                )

                else None
            )


            if (
                root_name
                not in
                cell118_local_names
            ):

                cell118_external_subscript_assignments.append({
                    "receiver":
                        root_name,

                    "line":
                        getattr(
                            node,
                            "lineno",
                            None,
                        ),
                })


# ============================================================
# 9) Resolve helper Runtime dependencies
# ============================================================

cell118_loaded_names = set()


if (
    cell118_function_node
    is not None
):

    for node in ast.walk(
        cell118_function_node
    ):

        if (
            isinstance(
                node,
                ast.Name,
            )

            and

            isinstance(
                node.ctx,
                ast.Load,
            )
        ):

            cell118_loaded_names.add(
                node.id
            )


cell118_runtime_dependency_names = sorted([
    name

    for name
    in cell118_loaded_names

    if (
        name
        not in
        cell118_local_names

        and

        name
        in globals()
    )
])


# ============================================================
# 10) Inspect current document support
# ============================================================

cell118_current_document_id = (
    cell117_candidate.get(
        "document_id"
    )
)


cell118_corpus_record = None


if (
    "corpus_target_index"
    in globals()

    and

    isinstance(
        corpus_target_index,
        dict,
    )
):

    cell118_corpus_record = (
        corpus_target_index.get(
            cell118_current_document_id
        )
    )


cell118_document_support_found = (
    isinstance(
        cell118_corpus_record,
        dict,
    )

    and

    cell118_corpus_record.get(
        "document_id"
    )
    ==
    cell118_current_document_id
)


# ============================================================
# 11) Compact key source lines
# ============================================================

CELL118_KEY_TERMS = [
    "corpus_target_index",
    "sentence_spans",
    "window_size",
    "candidates",
    "char_start",
    "char_end",
    "token_count",
    "winner",
    "sorted",
    "return",
    "raise",
]


cell118_key_source_lines = []


if isinstance(
    cell118_helper_source,
    str,
):

    for line_number, line in enumerate(
        cell118_helper_source.splitlines(),
        start=1,
    ):

        lower = (
            line.lower()
        )


        if any(
            term
            in
            lower

            for term
            in CELL118_KEY_TERMS
        ):

            cell118_key_source_lines.append({
                "line":
                    line_number,

                "text":
                    line.rstrip(),
            })


cell118_key_source_lines = (
    cell118_key_source_lines[
        :80
    ]
)


# ============================================================
# 12) Purity interpretation
# ============================================================

cell118_helper_source_available = (
    isinstance(
        cell118_helper_source,
        str,
    )

    and

    len(
        cell118_helper_source
    )
    >
    0
)


cell118_has_only_local_mutation = (
    len(
        cell118_external_mutation_calls
    )
    ==
    0

    and

    len(
        cell118_external_subscript_assignments
    )
    ==
    0
)


cell118_local_candidate_accumulation_confirmed = (
    any(
        item.get(
            "receiver"
        )
        ==
        "candidates"

        and

        item.get(
            "method"
        )
        ==
        "append"

        for item
        in cell118_local_mutation_calls
    )
)


if (
    cell118_helper_source_available
    and
    cell118_document_support_found
    and
    cell118_has_only_local_mutation
):

    cell118_status_detail = (
        "ACTIVE_LOCAL_SELECTOR_CONFIRMED_READ_ONLY_WITH_LOCAL_ACCUMULATION"
    )

    cell118_next_action = (
        "DRY_RUN_PURE_LOCAL_SELECTOR_AND_COMPARE_WITH_CURRENT_TARGET"
    )


else:

    cell118_status_detail = (
        "ACTIVE_LOCAL_SELECTOR_PURITY_NOT_CONFIRMED"
    )

    cell118_next_action = (
        "DO_NOT_EXECUTE_LOCAL_SELECTOR"
    )


# ============================================================
# 13) Result record
# ============================================================

cell118_inspection_record = {
    "inspection_version":
        "active_local_target_helper_inspection_v2",

    "method":
        cell117_method,

    "document_id":
        cell118_current_document_id,

    "helper_signature":
        cell118_helper_signature,

    "runtime_dependency_names":
        cell118_runtime_dependency_names,

    "local_names":
        sorted(
            cell118_local_names
        ),

    "local_mutation_calls":
        cell118_local_mutation_calls,

    "external_mutation_calls":
        cell118_external_mutation_calls,

    "external_subscript_assignments":
        cell118_external_subscript_assignments,

    "local_candidate_accumulation_confirmed":
        cell118_local_candidate_accumulation_confirmed,

    "has_only_local_mutation":
        cell118_has_only_local_mutation,

    "document_support_found":
        cell118_document_support_found,

    "selector_called":
        False,

    "target_mutated":
        False,

    "status":
        cell118_status_detail,

    "next_action":
        cell118_next_action,
}


cell118_inspection_sha256 = (
    cell118_sha(
        cell118_inspection_record
    )
)


# ============================================================
# 14) Freeze protected state AFTER inspection
# ============================================================

cell118_after = {
    "question":
        cell118_sha(
            cell97_question_draft_record
        ),

    "targets":
        cell118_sha(
            target_evidence_records
        ),

    "case_index":
        cell118_sha(
            target_evidence_by_case_id
        ),

    "task_index":
        cell118_sha(
            target_evidence_by_task_draft_id
        ),

    "queue":
        cell118_sha(
            cell95_remaining_pilot_queue
        ),

    "pilot":
        cell118_sha({
            "packets":
                question_drafting_pilot_packets,

            "by_id":
                question_drafting_pilot_by_id,
        }),
}


# ============================================================
# 15) Gates
# ============================================================

cell118_gates = {
    "cell117_is_pass":
        globals().get(
            "cell_117_pass"
        )
        is True,

    "required_objects_exist":
        len(
            cell118_missing_objects
        )
        ==
        0,

    "helper_signature_resolved":
        cell118_helper_signature
        is not None,

    "helper_source_resolved":
        cell118_helper_source_available,

    "helper_ast_parsed":
        cell118_function_node
        is not None,

    "current_document_id_resolved":
        isinstance(
            cell118_current_document_id,
            str,
        ),

    "document_support_found":
        cell118_document_support_found,

    "local_candidate_accumulation_confirmed":
        cell118_local_candidate_accumulation_confirmed,

    "no_external_mutating_method_calls":
        len(
            cell118_external_mutation_calls
        )
        ==
        0,

    "no_external_subscript_assignments":
        len(
            cell118_external_subscript_assignments
        )
        ==
        0,

    "helper_has_only_local_mutation":
        cell118_has_only_local_mutation,

    "selector_not_called":
        cell118_inspection_record[
            "selector_called"
        ]
        is False,

    "protected_state_byte_stable":
        cell118_before
        ==
        cell118_after,

    "no_artifact_write":
        True,
}


cell_118_pass = all(
    cell118_gates.values()
)


cell118_status = (
    "ACTIVE_LOCAL_TARGET_HELPER_INSPECTED"

    if cell_118_pass

    else
    "ACTIVE_LOCAL_TARGET_HELPER_INSPECTION_FAILED"
)


# ============================================================
# 16) Diagnostics
# ============================================================

print(
    "=== ACTIVE LOCAL TARGET HELPER — PURITY INSPECTION ==="
)


print(
    "\nHelper signature:",
    cell118_helper_signature
)

print(
    "Current document:",
    cell118_current_document_id
)


print(
    "\n=== LOCAL SCOPE ==="
)

print(
    "Local names:",
    sorted(
        cell118_local_names
    )
)


print(
    "\n=== LOCAL MUTATION CALLS ==="
)

print(
    cell118_local_mutation_calls
)


print(
    "\n=== EXTERNAL MUTATION CALLS ==="
)

print(
    cell118_external_mutation_calls
)


print(
    "\n=== EXTERNAL SUBSCRIPT ASSIGNMENTS ==="
)

print(
    cell118_external_subscript_assignments
)


print(
    "\n=== RUNTIME DEPENDENCIES ==="
)

print(
    cell118_runtime_dependency_names
)


print(
    "\n=== HELPER KEY SOURCE LINES ==="
)

for item in (
    cell118_key_source_lines
):

    print(
        f"{item['line']:04d}:",
        item[
            "text"
        ]
    )


print(
    "\n=== PURITY INTERPRETATION ==="
)

print(
    "Local candidates.append confirmed:",
    cell118_local_candidate_accumulation_confirmed
)

print(
    "Has only local mutation:",
    cell118_has_only_local_mutation
)

print(
    "Document support found:",
    cell118_document_support_found
)

print(
    "Status:",
    cell118_status_detail
)

print(
    "Next action:",
    cell118_next_action
)

print(
    "Inspection SHA256:",
    cell118_inspection_sha256
)


print(
    "\n=== FAILURES ==="
)

print(
    "Prerequisite failures:",
    cell118_prerequisite_failures
)

print(
    "Inspection failures:",
    cell118_inspection_failures
)


print(
    "\nSTATUS:",
    cell118_status
)


print(
    "\n=== CODE CELL 118 GATES ==="
)


for name, passed in (
    cell118_gates.items()
):

    print(
        name,
        ":",
        (
            "PASS"
            if passed
            else "FAIL"
        ),
    )


print(
    "\nFAILED GATES:",
    [
        name

        for name, passed
        in cell118_gates.items()

        if not passed
    ],
)


print(
    "\nCODE CELL 118 GATE:",
    (
        "PASS"
        if cell_118_pass
        else "FAIL"
    ),
)


assert cell_118_pass, (
    "Active local Target helper purity inspection failed. "
    "Do not execute the selector or modify Target Evidence. "
    "Repair this SAME Code Cell 118."
)
```

```python
# Code Cell 119

# ============================================================
# Dry-run the confirmed pure local selector and compare its
# output with the CURRENT stored Target Evidence.
#
# The helper is allowed to execute because Cell 118 proved
# that it performs no external Runtime mutation.
#
# No Target registration.
# No Target ID creation.
# No question mutation.
# No queue mutation.
# No artifact write.
# ============================================================


# ============================================================
# 1) Prerequisites
# ============================================================

cell119_prerequisite_failures = []


if globals().get(
    "cell_118_pass"
) is not True:

    cell119_prerequisite_failures.append(
        "CELL118_NOT_PASS"
    )


if (
    globals().get(
        "cell118_status"
    )
    !=
    "ACTIVE_LOCAL_TARGET_HELPER_INSPECTED"
):

    cell119_prerequisite_failures.append({
        "reason":
            "CELL118_STATUS_UNEXPECTED",

        "actual":
            globals().get(
                "cell118_status"
            ),
    })


CELL119_REQUIRED_OBJECTS = [
    "select_website_local_target",
    "cell117_candidate",
    "cell97_question_draft_record",
    "target_evidence_records",
    "target_evidence_by_case_id",
    "target_evidence_by_task_draft_id",
    "cell95_remaining_pilot_queue",
    "question_drafting_pilot_packets",
    "question_drafting_pilot_by_id",
]


cell119_missing_objects = [
    name

    for name
    in CELL119_REQUIRED_OBJECTS

    if name not in globals()
]


if cell119_missing_objects:

    cell119_prerequisite_failures.append({
        "reason":
            "REQUIRED_RUNTIME_OBJECTS_MISSING",

        "objects":
            cell119_missing_objects,
    })


# ============================================================
# 2) Helper
# ============================================================

def cell119_sha(value):

    return hashlib.sha256(
        stable_json_key(
            value
        ).encode(
            "utf-8"
        )
    ).hexdigest()


# ============================================================
# 3) Resolve current identity and stored Target
# ============================================================

cell119_document_id = (
    cell117_candidate.get(
        "document_id"
    )
)


cell119_task_draft_id = (
    cell97_question_draft_record.get(
        "task_draft_id"
    )
)


cell119_current_target_id = (
    cell97_question_draft_record.get(
        "target_evidence_id"
    )
)


cell119_current_target_record = (
    target_evidence_by_task_draft_id.get(
        cell119_task_draft_id
    )
)


cell119_current_target_parts = (
    cell119_current_target_record.get(
        "target_parts",
        [],
    )

    if isinstance(
        cell119_current_target_record,
        dict,
    )

    else []
)


# ============================================================
# 4) Freeze protected state BEFORE Dry Run
# ============================================================

cell119_before = {
    "question":
        cell119_sha(
            cell97_question_draft_record
        ),

    "targets":
        cell119_sha(
            target_evidence_records
        ),

    "case_index":
        cell119_sha(
            target_evidence_by_case_id
        ),

    "task_index":
        cell119_sha(
            target_evidence_by_task_draft_id
        ),

    "queue":
        cell119_sha(
            cell95_remaining_pilot_queue
        ),

    "pilot":
        cell119_sha({
            "packets":
                question_drafting_pilot_packets,

            "by_id":
                question_drafting_pilot_by_id,
        }),
}


# ============================================================
# 5) Execute PURE helper Dry Run
# ============================================================

cell119_dry_run_failures = []


try:

    cell119_dry_run_parts = (
        select_website_local_target(
            cell119_document_id
        )
    )


except Exception as exc:

    cell119_dry_run_parts = None

    cell119_dry_run_failures.append({
        "reason":
            "PURE_LOCAL_SELECTOR_DRY_RUN_FAILED",

        "error":
            repr(
                exc
            ),
    })


# ============================================================
# 6) Compare Dry Run with current stored Target
# ============================================================

cell119_dry_run_is_list = (
    isinstance(
        cell119_dry_run_parts,
        list,
    )
)


cell119_current_parts_is_list = (
    isinstance(
        cell119_current_target_parts,
        list,
    )
)


cell119_part_count_match = (
    cell119_dry_run_is_list

    and

    cell119_current_parts_is_list

    and

    len(
        cell119_dry_run_parts
    )
    ==
    len(
        cell119_current_target_parts
    )
)


cell119_exact_parts_match = (
    cell119_part_count_match

    and

    cell119_sha(
        cell119_dry_run_parts
    )
    ==
    cell119_sha(
        cell119_current_target_parts
    )
)


CELL119_COMPARE_FIELDS = [
    "document_id",
    "role",
    "char_start",
    "char_end",
    "char_count",
    "token_count",
    "target_text_sha256",
]


cell119_field_comparison = []


if (
    cell119_dry_run_is_list
    and
    cell119_current_parts_is_list
):

    for index in range(
        min(
            len(
                cell119_dry_run_parts
            ),
            len(
                cell119_current_target_parts
            ),
        )
    ):

        dry_part = (
            cell119_dry_run_parts[
                index
            ]
        )

        stored_part = (
            cell119_current_target_parts[
                index
            ]
        )


        comparison = {
            field: (
                dry_part.get(
                    field
                )
                ==
                stored_part.get(
                    field
                )
            )

            for field
            in CELL119_COMPARE_FIELDS
        }


        cell119_field_comparison.append({
            "part_index":
                index,

            "fields":
                comparison,

            "all_match":
                all(
                    comparison.values()
                ),
        })


cell119_semantic_span_match = (
    cell119_part_count_match

    and

    all(
        item[
            "all_match"
        ]

        for item
        in cell119_field_comparison
    )
)


# ============================================================
# 7) Freeze protected state AFTER Dry Run
# ============================================================

cell119_after = {
    "question":
        cell119_sha(
            cell97_question_draft_record
        ),

    "targets":
        cell119_sha(
            target_evidence_records
        ),

    "case_index":
        cell119_sha(
            target_evidence_by_case_id
        ),

    "task_index":
        cell119_sha(
            target_evidence_by_task_draft_id
        ),

    "queue":
        cell119_sha(
            cell95_remaining_pilot_queue
        ),

    "pilot":
        cell119_sha({
            "packets":
                question_drafting_pilot_packets,

            "by_id":
                question_drafting_pilot_by_id,
        }),
}


cell119_protected_state_stable = (
    cell119_before
    ==
    cell119_after
)


# ============================================================
# 8) Interpretation
# ============================================================

if (
    len(
        cell119_dry_run_failures
    )
    ==
    0

    and

    cell119_semantic_span_match
):

    cell119_dry_run_status = (
        "LOCAL_SELECTOR_DETERMINISTICALLY_REPRODUCES_CURRENT_TARGET"
    )

    cell119_next_action = (
        "INSPECT_ALL_LOCAL_WINDOW_CANDIDATES_BEFORE_DECIDING_"
        "WHETHER_SAFE_RETARGET_IS_POSSIBLE"
    )


elif (
    len(
        cell119_dry_run_failures
    )
    ==
    0

    and

    not cell119_semantic_span_match
):

    cell119_dry_run_status = (
        "LOCAL_SELECTOR_OUTPUT_DIFFERS_FROM_CURRENT_STORED_TARGET"
    )

    cell119_next_action = (
        "INSPECT_SELECTOR_DRIFT_BEFORE_ANY_RETARGET"
    )


else:

    cell119_dry_run_status = (
        "LOCAL_SELECTOR_DRY_RUN_UNRESOLVED"
    )

    cell119_next_action = (
        "DO_NOT_RETARGET"
    )


# ============================================================
# 9) Result
# ============================================================

cell119_local_selector_dry_run = {
    "inspection_version":
        "local_selector_dry_run_v1",

    "document_id":
        cell119_document_id,

    "task_draft_id":
        cell119_task_draft_id,

    "current_target_evidence_id":
        cell119_current_target_id,

    "dry_run_part_count":
        (
            len(
                cell119_dry_run_parts
            )

            if cell119_dry_run_is_list

            else None
        ),

    "stored_part_count":
        (
            len(
                cell119_current_target_parts
            )

            if cell119_current_parts_is_list

            else None
        ),

    "exact_parts_match":
        cell119_exact_parts_match,

    "semantic_span_match":
        cell119_semantic_span_match,

    "protected_state_stable":
        cell119_protected_state_stable,

    "target_registered":
        False,

    "target_id_created":
        False,

    "status":
        cell119_dry_run_status,

    "next_action":
        cell119_next_action,
}


cell119_local_selector_dry_run_sha256 = (
    cell119_sha(
        cell119_local_selector_dry_run
    )
)


# ============================================================
# 10) Gates
# ============================================================

cell119_gates = {
    "cell118_is_pass":
        globals().get(
            "cell_118_pass"
        )
        is True,

    "required_objects_exist":
        len(
            cell119_missing_objects
        )
        ==
        0,

    "current_document_resolved":
        isinstance(
            cell119_document_id,
            str,
        ),

    "current_target_record_resolved":
        isinstance(
            cell119_current_target_record,
            dict,
        ),

    "dry_run_succeeded":
        len(
            cell119_dry_run_failures
        )
        ==
        0,

    "dry_run_returned_parts":
        (
            cell119_dry_run_is_list

            and

            len(
                cell119_dry_run_parts
            )
            >
            0
        ),

    "stored_target_has_parts":
        (
            cell119_current_parts_is_list

            and

            len(
                cell119_current_target_parts
            )
            >
            0
        ),

    "protected_state_remains_byte_stable":
        cell119_protected_state_stable,

    "target_not_registered":
        cell119_local_selector_dry_run[
            "target_registered"
        ]
        is False,

    "target_id_not_created":
        cell119_local_selector_dry_run[
            "target_id_created"
        ]
        is False,

    "comparison_completed":
        isinstance(
            cell119_semantic_span_match,
            bool,
        ),

    "no_artifact_write":
        True,
}


cell_119_pass = all(
    cell119_gates.values()
)


cell119_status = (
    "PURE_LOCAL_SELECTOR_DRY_RUN_COMPLETED"

    if cell_119_pass

    else
    "PURE_LOCAL_SELECTOR_DRY_RUN_FAILED"
)


# ============================================================
# 11) Compact diagnostics
# ============================================================

print(
    "=== PURE LOCAL SELECTOR DRY RUN ==="
)


print(
    "\nDocument:",
    cell119_document_id
)

print(
    "Current Target ID:",
    cell119_current_target_id
)


print(
    "\n=== STORED TARGET PARTS ==="
)

for part in (
    cell119_current_target_parts
):

    print({
        key:
            part.get(
                key
            )

        for key
        in CELL119_COMPARE_FIELDS
    })


print(
    "\n=== DRY-RUN TARGET PARTS ==="
)

if cell119_dry_run_is_list:

    for part in (
        cell119_dry_run_parts
    ):

        print({
            key:
                part.get(
                    key
                )

            for key
            in CELL119_COMPARE_FIELDS
        })


print(
    "\n=== COMPARISON ==="
)

print(
    "Part count match:",
    cell119_part_count_match
)

print(
    "Exact full-parts match:",
    cell119_exact_parts_match
)

print(
    "Semantic/span fields match:",
    cell119_semantic_span_match
)

print(
    "Field comparison:",
    cell119_field_comparison
)


print(
    "\n=== STATE SAFETY ==="
)

print(
    "Protected state stable:",
    cell119_protected_state_stable
)


print(
    "\n=== INTERPRETATION ==="
)

print(
    "Dry-run status:",
    cell119_dry_run_status
)

print(
    "Next action:",
    cell119_next_action
)

print(
    "Inspection SHA256:",
    cell119_local_selector_dry_run_sha256
)


print(
    "\n=== FAILURES ==="
)

print(
    "Prerequisite failures:",
    cell119_prerequisite_failures
)

print(
    "Dry-run failures:",
    cell119_dry_run_failures
)


print(
    "\nSTATUS:",
    cell119_status
)


print(
    "\n=== CODE CELL 119 GATES ==="
)


for name, passed in (
    cell119_gates.items()
):

    print(
        name,
        ":",
        (
            "PASS"
            if passed
            else "FAIL"
        ),
    )


print(
    "\nFAILED GATES:",
    [
        name

        for name, passed
        in cell119_gates.items()

        if not passed
    ],
)


print(
    "\nCODE CELL 119 GATE:",
    (
        "PASS"
        if cell_119_pass
        else "FAIL"
    ),
)


assert cell_119_pass, (
    "Pure local selector Dry Run failed. "
    "Do not retarget or modify Target Evidence. "
    "Repair this SAME Code Cell 119."
)
```

<h3>Repair همان Cell 120 — استخراج Schema واقعی Sentence Span از خود Helper</h3>

<p>
FAIL قبلی به این دلیل بود که Cell 120 فرض کرده بود هر Sentence Span یک
<code>dict</code>
با Fieldهای
<code>char_start</code>
و
<code>char_end</code>
است.
</p>

<p>
در این Repair هیچ Schema حدس زده نمی‌شود.
Source واقعی
<code>select_website_local_target</code>
با AST بررسی می‌شود تا دقیقاً مشخص شود Helper برای
<code>left</code>
و
<code>right</code>
از چه Key یا Indexی برای Start و End استفاده می‌کند.
</p>

<p>
سپس تمام Windowها با همان Accessor واقعی Helper بازسازی می‌شوند و Target فعلی و Alternativeهای احتمالی دوباره بررسی می‌شوند.
</p>

<p>
هیچ Target جدیدی ثبت نمی‌شود و هیچ State تغییر نمی‌کند.
</p>

```python
# Code Cell 120

# ============================================================
# SAME-CELL REPAIR
#
# Enumerate all local windows using the ACTUAL sentence-span
# access pattern extracted from select_website_local_target.
#
# No span schema guessing.
# No Target registration.
# No Target ID creation.
# No state mutation.
# No artifact write.
# ============================================================

import ast
import textwrap


# ============================================================
# 1) Prerequisites
# ============================================================

cell120_prerequisite_failures = []


if globals().get(
    "cell_119_pass"
) is not True:

    cell120_prerequisite_failures.append(
        "CELL119_NOT_PASS"
    )


if (
    globals().get(
        "cell119_status"
    )
    !=
    "PURE_LOCAL_SELECTOR_DRY_RUN_COMPLETED"
):

    cell120_prerequisite_failures.append({
        "reason":
            "CELL119_STATUS_UNEXPECTED",

        "actual":
            globals().get(
                "cell119_status"
            ),
    })


CELL120_REQUIRED_OBJECTS = [
    "cell118_helper_source",
    "cell119_document_id",
    "cell119_current_target_parts",
    "corpus_target_index",
    "nb11_sentence_spans_with_offsets",
    "nb11_token_count",
    "cell97_question_draft_record",
    "target_evidence_records",
    "target_evidence_by_case_id",
    "target_evidence_by_task_draft_id",
    "cell95_remaining_pilot_queue",
    "question_drafting_pilot_packets",
    "question_drafting_pilot_by_id",
]


cell120_missing_objects = [
    name

    for name
    in CELL120_REQUIRED_OBJECTS

    if name not in globals()
]


if cell120_missing_objects:

    cell120_prerequisite_failures.append({
        "reason":
            "REQUIRED_RUNTIME_OBJECTS_MISSING",

        "objects":
            cell120_missing_objects,
    })


# ============================================================
# 2) Helper
# ============================================================

def cell120_sha(value):

    return hashlib.sha256(
        stable_json_key(
            value
        ).encode(
            "utf-8"
        )
    ).hexdigest()


def cell120_get_subscript_key(node):

    if not isinstance(
        node,
        ast.Subscript,
    ):
        return None

    slice_node = (
        node.slice
    )

    if isinstance(
        slice_node,
        ast.Constant,
    ):
        return (
            slice_node.value
        )

    return None


def cell120_extract_accessor(
    tree,
    assigned_name,
    root_name,
):

    matches = []


    for node in ast.walk(
        tree
    ):

        if not isinstance(
            node,
            ast.Assign,
        ):
            continue


        if not any(
            isinstance(
                target,
                ast.Name,
            )

            and

            target.id
            ==
            assigned_name

            for target
            in node.targets
        ):
            continue


        for child in ast.walk(
            node.value
        ):

            if not isinstance(
                child,
                ast.Subscript,
            ):
                continue


            if not (
                isinstance(
                    child.value,
                    ast.Name,
                )

                and

                child.value.id
                ==
                root_name
            ):
                continue


            matches.append(
                cell120_get_subscript_key(
                    child
                )
            )


    matches = [
        item

        for item
        in matches

        if item is not None
    ]


    return (
        matches[
            0
        ]

        if len(
            matches
        )
        ==
        1

        else None
    )


def cell120_apply_accessor(
    item,
    accessor,
):

    if isinstance(
        accessor,
        str,
    ):

        assert isinstance(
            item,
            dict,
        ), (
            "Helper uses a string-key accessor, "
            "but sentence span item is not a dict."
        )

        return item[
            accessor
        ]


    if isinstance(
        accessor,
        int,
    ):

        assert isinstance(
            item,
            (list, tuple),
        ), (
            "Helper uses an integer-index accessor, "
            "but sentence span item is not a list/tuple."
        )

        return item[
            accessor
        ]


    raise AssertionError(
        f"Unsupported span accessor: {accessor!r}"
    )


# ============================================================
# 3) Freeze protected state BEFORE inspection
# ============================================================

cell120_before = {
    "question":
        cell120_sha(
            cell97_question_draft_record
        ),

    "targets":
        cell120_sha(
            target_evidence_records
        ),

    "case_index":
        cell120_sha(
            target_evidence_by_case_id
        ),

    "task_index":
        cell120_sha(
            target_evidence_by_task_draft_id
        ),

    "queue":
        cell120_sha(
            cell95_remaining_pilot_queue
        ),

    "pilot":
        cell120_sha({
            "packets":
                question_drafting_pilot_packets,

            "by_id":
                question_drafting_pilot_by_id,
        }),
}


# ============================================================
# 4) Resolve current document
# ============================================================

cell120_document_record = (
    corpus_target_index.get(
        cell119_document_id
    )
)


cell120_document_text = (
    cell120_document_record.get(
        "text"
    )

    if isinstance(
        cell120_document_record,
        dict,
    )

    else None
)


# ============================================================
# 5) Generate actual sentence spans
# ============================================================

cell120_failures = []


try:

    cell120_sentence_spans = (
        nb11_sentence_spans_with_offsets(
            cell120_document_text
        )
    )


except Exception as exc:

    cell120_sentence_spans = None

    cell120_failures.append({
        "reason":
            "SENTENCE_SPAN_GENERATION_FAILED",

        "error":
            repr(
                exc
            ),
    })


cell120_sentence_spans_exist = (
    isinstance(
        cell120_sentence_spans,
        list,
    )

    and

    len(
        cell120_sentence_spans
    )
    >
    0
)


# ============================================================
# 6) Parse actual helper source
# ============================================================

cell120_helper_tree = None
cell120_helper_source_dedented = None


try:

    cell120_helper_source_dedented = (
        textwrap.dedent(
            cell118_helper_source
        )
    )


    cell120_helper_tree = ast.parse(
        cell120_helper_source_dedented
    )


except Exception as exc:

    cell120_failures.append({
        "reason":
            "HELPER_AST_PARSE_FAILED",

        "error":
            repr(
                exc
            ),
    })


# ============================================================
# 7) Resolve window_size from actual Helper
# ============================================================

cell120_window_size_expression = None
cell120_window_size = None


if (
    cell120_helper_tree
    is not None

    and

    cell120_sentence_spans_exist
):

    try:

        cell120_window_assignments = []


        for node in ast.walk(
            cell120_helper_tree
        ):

            if not isinstance(
                node,
                ast.Assign,
            ):
                continue


            if any(
                isinstance(
                    target,
                    ast.Name,
                )

                and

                target.id
                ==
                "window_size"

                for target
                in node.targets
            ):

                cell120_window_assignments.append(
                    node
                )


        assert (
            len(
                cell120_window_assignments
            )
            ==
            1
        )


        cell120_window_node = (
            cell120_window_assignments[
                0
            ].value
        )


        cell120_window_size_expression = (
            ast.get_source_segment(
                cell120_helper_source_dedented,
                cell120_window_node,
            )
        )


        cell120_window_size = eval(
            compile(
                ast.Expression(
                    body=
                        cell120_window_node
                ),
                "<cell120-window-size>",
                "eval",
            ),
            {
                "__builtins__":
                    {},

                "min":
                    min,

                "len":
                    len,
            },
            {
                "sentence_spans":
                    cell120_sentence_spans,
            },
        )


    except Exception as exc:

        cell120_failures.append({
            "reason":
                "WINDOW_SIZE_RESOLUTION_FAILED",

            "error":
                repr(
                    exc
                ),
        })


cell120_window_size_valid = (
    isinstance(
        cell120_window_size,
        int,
    )

    and

    cell120_window_size
    >
    0

    and

    cell120_sentence_spans_exist

    and

    cell120_window_size
    <=
    len(
        cell120_sentence_spans
    )
)


# ============================================================
# 8) Resolve actual left/right span accessors from Helper AST
# ============================================================

cell120_left_start_accessor = None
cell120_right_end_accessor = None


if (
    cell120_helper_tree
    is not None
):

    cell120_left_start_accessor = (
        cell120_extract_accessor(
            cell120_helper_tree,
            "char_start",
            "left",
        )
    )


    cell120_right_end_accessor = (
        cell120_extract_accessor(
            cell120_helper_tree,
            "char_end",
            "right",
        )
    )


cell120_accessors_resolved = (
    cell120_left_start_accessor
    is not None

    and

    cell120_right_end_accessor
    is not None
)


# ============================================================
# 9) Validate accessors against actual span sample
# ============================================================

cell120_span_sample = (
    cell120_sentence_spans[
        :3
    ]

    if cell120_sentence_spans_exist

    else []
)


cell120_accessor_validation_pass = False


if (
    cell120_sentence_spans_exist
    and
    cell120_accessors_resolved
):

    try:

        cell120_sample_start = (
            cell120_apply_accessor(
                cell120_sentence_spans[
                    0
                ],
                cell120_left_start_accessor,
            )
        )


        cell120_sample_end = (
            cell120_apply_accessor(
                cell120_sentence_spans[
                    0
                ],
                cell120_right_end_accessor,
            )
        )


        cell120_accessor_validation_pass = (
            isinstance(
                cell120_sample_start,
                int,
            )

            and

            isinstance(
                cell120_sample_end,
                int,
            )

            and

            cell120_sample_start
            <=
            cell120_sample_end
        )


    except Exception as exc:

        cell120_failures.append({
            "reason":
                "SPAN_ACCESSOR_VALIDATION_FAILED",

            "error":
                repr(
                    exc
                ),
        })


# ============================================================
# 10) Enumerate exact consecutive windows
# ============================================================

cell120_candidates = []


if (
    cell120_window_size_valid
    and
    cell120_accessor_validation_pass
):

    try:

        for index in range(
            len(
                cell120_sentence_spans
            )
            -
            cell120_window_size
            +
            1
        ):

            left = (
                cell120_sentence_spans[
                    index
                ]
            )


            right = (
                cell120_sentence_spans[
                    index
                    +
                    cell120_window_size
                    -
                    1
                ]
            )


            char_start = (
                cell120_apply_accessor(
                    left,
                    cell120_left_start_accessor,
                )
            )


            char_end = (
                cell120_apply_accessor(
                    right,
                    cell120_right_end_accessor,
                )
            )


            target_text = (
                cell120_document_text[
                    char_start:
                    char_end
                ]
            )


            token_count = (
                nb11_token_count(
                    target_text
                )
            )


            cell120_candidates.append({
                "window_index":
                    index,

                "char_start":
                    char_start,

                "char_end":
                    char_end,

                "char_count":
                    char_end
                    -
                    char_start,

                "token_count":
                    token_count,

                "target_text":
                    target_text,

                "target_text_sha256":
                    hashlib.sha256(
                        target_text.encode(
                            "utf-8"
                        )
                    ).hexdigest(),
            })


    except Exception as exc:

        cell120_failures.append({
            "reason":
                "WINDOW_ENUMERATION_FAILED",

            "error":
                repr(
                    exc
                ),
        })


# ============================================================
# 11) Resolve current stored Target span
# ============================================================

cell120_current_part = (
    cell119_current_target_parts[
        0
    ]

    if (
        isinstance(
            cell119_current_target_parts,
            list,
        )

        and

        len(
            cell119_current_target_parts
        )
        ==
        1
    )

    else None
)


cell120_current_span = (
    (
        cell120_current_part.get(
            "char_start"
        ),
        cell120_current_part.get(
            "char_end"
        ),
    )

    if isinstance(
        cell120_current_part,
        dict,
    )

    else None
)


# ============================================================
# 12) Evidence markers
# ============================================================

CELL120_CORE_LIST_ANCHORS = [
    "خرج",
    "پس‌انداز",
    "اعتبار",
    "بدهی",
    "سرمایه‌گذاری",
    "مدیریت ریسک",
    "تصمیم‌گیری مالی",
]


CELL120_TOTAL_COUNT_MARKER = (
    "چهارده مفهوم سواد مالی"
)


CELL120_HUMAN_OBJECTION_MARKERS = [
    "شوک‌های درآمدی",
    "هزین‌های",
]


# ============================================================
# 13) Classify candidates
# ============================================================

cell120_candidate_summaries = []


for candidate in (
    cell120_candidates
):

    text = (
        candidate[
            "target_text"
        ]
    )


    core_hits = [
        anchor

        for anchor
        in CELL120_CORE_LIST_ANCHORS

        if anchor in text
    ]


    objection_hits = [
        marker

        for marker
        in CELL120_HUMAN_OBJECTION_MARKERS

        if marker in text
    ]


    has_total_count = (
        CELL120_TOTAL_COUNT_MARKER
        in
        text
    )


    has_full_core_list = (
        len(
            core_hits
        )
        ==
        len(
            CELL120_CORE_LIST_ANCHORS
        )
    )


    is_current_span = (
        (
            candidate[
                "char_start"
            ],
            candidate[
                "char_end"
            ],
        )
        ==
        cell120_current_span
    )


    strict_question_support = (
        has_total_count
        and
        has_full_core_list
    )


    strict_safe_alternative = (
        strict_question_support
        and
        len(
            objection_hits
        )
        ==
        0
        and
        not is_current_span
    )


    cell120_candidate_summaries.append({
        "window_index":
            candidate[
                "window_index"
            ],

        "char_start":
            candidate[
                "char_start"
            ],

        "char_end":
            candidate[
                "char_end"
            ],

        "token_count":
            candidate[
                "token_count"
            ],

        "is_current_span":
            is_current_span,

        "core_anchor_count":
            len(
                core_hits
            ),

        "has_total_count":
            has_total_count,

        "objection_hits":
            objection_hits,

        "strict_question_support":
            strict_question_support,

        "strict_safe_alternative":
            strict_safe_alternative,

        "preview":
            " ".join(
                text.split()
            )[
                :220
            ],
    })


# ============================================================
# 14) Relevant subsets
# ============================================================

cell120_current_matches = [
    item

    for item
    in cell120_candidate_summaries

    if item[
        "is_current_span"
    ]
]


cell120_question_support_candidates = [
    item

    for item
    in cell120_candidate_summaries

    if item[
        "strict_question_support"
    ]
]


cell120_safe_alternatives = [
    item

    for item
    in cell120_candidate_summaries

    if item[
        "strict_safe_alternative"
    ]
]


cell120_current_span_reproduced = (
    len(
        cell120_current_matches
    )
    ==
    1
)


cell120_safe_alternative_exists = (
    len(
        cell120_safe_alternatives
    )
    >
    0
)


# ============================================================
# 15) Interpretation
# ============================================================

if (
    cell120_current_span_reproduced
    and
    cell120_safe_alternative_exists
):

    cell120_status_detail = (
        "SAFE_LOCAL_WINDOW_ALTERNATIVE_CANDIDATE_FOUND"
    )

    cell120_next_action = (
        "INSPECT_SAFE_LOCAL_ALTERNATIVE_WITH_FULL_PROVENANCE_BEFORE_RETARGET"
    )


elif (
    cell120_current_span_reproduced
    and
    not cell120_safe_alternative_exists
):

    cell120_status_detail = (
        "NO_SAFE_LOCAL_WINDOW_ALTERNATIVE_FOUND"
    )

    cell120_next_action = (
        "KEEP_PILOT_BLOCKED_AND_INSPECT_PILOT_CASE_REPLACEMENT_POLICY"
    )


else:

    cell120_status_detail = (
        "LOCAL_WINDOW_ENUMERATION_NOT_CONCLUSIVE"
    )

    cell120_next_action = (
        "DO_NOT_RETARGET"
    )


# ============================================================
# 16) Result
# ============================================================

cell120_local_window_audit = {
    "inspection_version":
        "current_document_local_window_audit_v2",

    "document_id":
        cell119_document_id,

    "sentence_span_item_type":
        (
            type(
                cell120_sentence_spans[
                    0
                ]
            ).__name__

            if cell120_sentence_spans_exist

            else None
        ),

    "left_start_accessor":
        cell120_left_start_accessor,

    "right_end_accessor":
        cell120_right_end_accessor,

    "window_size_expression":
        cell120_window_size_expression,

    "window_size":
        cell120_window_size,

    "sentence_count":
        (
            len(
                cell120_sentence_spans
            )

            if cell120_sentence_spans_exist

            else None
        ),

    "candidate_count":
        len(
            cell120_candidates
        ),

    "question_support_candidate_count":
        len(
            cell120_question_support_candidates
        ),

    "safe_alternative_count":
        len(
            cell120_safe_alternatives
        ),

    "current_span_reproduced":
        cell120_current_span_reproduced,

    "safe_alternative_exists":
        cell120_safe_alternative_exists,

    "target_registered":
        False,

    "target_id_created":
        False,

    "status":
        cell120_status_detail,

    "next_action":
        cell120_next_action,
}


cell120_local_window_audit_sha256 = (
    cell120_sha(
        cell120_local_window_audit
    )
)


# ============================================================
# 17) Freeze protected state AFTER inspection
# ============================================================

cell120_after = {
    "question":
        cell120_sha(
            cell97_question_draft_record
        ),

    "targets":
        cell120_sha(
            target_evidence_records
        ),

    "case_index":
        cell120_sha(
            target_evidence_by_case_id
        ),

    "task_index":
        cell120_sha(
            target_evidence_by_task_draft_id
        ),

    "queue":
        cell120_sha(
            cell95_remaining_pilot_queue
        ),

    "pilot":
        cell120_sha({
            "packets":
                question_drafting_pilot_packets,

            "by_id":
                question_drafting_pilot_by_id,
        }),
}


# ============================================================
# 18) Gates
# ============================================================

cell120_gates = {
    "cell119_is_pass":
        globals().get(
            "cell_119_pass"
        )
        is True,

    "required_objects_exist":
        len(
            cell120_missing_objects
        )
        ==
        0,

    "document_record_resolved":
        isinstance(
            cell120_document_record,
            dict,
        ),

    "document_text_resolved":
        isinstance(
            cell120_document_text,
            str,
        ),

    "sentence_spans_exist":
        cell120_sentence_spans_exist,

    "helper_ast_parsed":
        cell120_helper_tree
        is not None,

    "window_size_resolved":
        cell120_window_size_valid,

    "span_accessors_resolved_from_helper":
        cell120_accessors_resolved,

    "span_accessors_validate_on_real_data":
        cell120_accessor_validation_pass,

    "candidate_enumeration_succeeded":
        (
            len(
                cell120_failures
            )
            ==
            0

            and

            len(
                cell120_candidates
            )
            >
            0
        ),

    "current_span_reproduced_once":
        cell120_current_span_reproduced,

    "safe_alternative_decision_completed":
        isinstance(
            cell120_safe_alternative_exists,
            bool,
        ),

    "target_not_registered":
        cell120_local_window_audit[
            "target_registered"
        ]
        is False,

    "target_id_not_created":
        cell120_local_window_audit[
            "target_id_created"
        ]
        is False,

    "protected_state_byte_stable":
        cell120_before
        ==
        cell120_after,

    "no_artifact_write":
        True,
}


cell_120_pass = all(
    cell120_gates.values()
)


cell120_status = (
    "CURRENT_DOCUMENT_LOCAL_WINDOWS_INSPECTED"

    if cell_120_pass

    else
    "CURRENT_DOCUMENT_LOCAL_WINDOW_INSPECTION_FAILED"
)


# ============================================================
# 19) Compact diagnostics
# ============================================================

print(
    "=== CURRENT DOCUMENT LOCAL WINDOW AUDIT ==="
)


print(
    "\n=== REAL SPAN SCHEMA ==="
)

print(
    "Span item type:",
    (
        type(
            cell120_sentence_spans[
                0
            ]
        ).__name__

        if cell120_sentence_spans_exist

        else None
    )
)

print(
    "Span sample:",
    cell120_span_sample
)

print(
    "Left start accessor:",
    cell120_left_start_accessor
)

print(
    "Right end accessor:",
    cell120_right_end_accessor
)


print(
    "\n=== WINDOW MODEL ==="
)

print(
    "Window-size expression:",
    cell120_window_size_expression
)

print(
    "Window size:",
    cell120_window_size
)

print(
    "Sentence count:",
    (
        len(
            cell120_sentence_spans
        )

        if cell120_sentence_spans_exist

        else None
    )
)

print(
    "Candidate count:",
    len(
        cell120_candidates
    )
)


print(
    "\n=== CURRENT TARGET ==="
)

print(
    "Stored span:",
    cell120_current_span
)

print(
    "Reproduced exactly once:",
    cell120_current_span_reproduced
)


print(
    "\n=== QUESTION-SUPPORTING WINDOWS ==="
)

print(
    "Count:",
    len(
        cell120_question_support_candidates
    )
)


for item in (
    cell120_question_support_candidates[
        :10
    ]
):

    print({
        "window":
            item[
                "window_index"
            ],

        "span":
            (
                item[
                    "char_start"
                ],
                item[
                    "char_end"
                ],
            ),

        "tokens":
            item[
                "token_count"
            ],

        "current":
            item[
                "is_current_span"
            ],

        "objection_hits":
            item[
                "objection_hits"
            ],

        "safe":
            item[
                "strict_safe_alternative"
            ],

        "preview":
            item[
                "preview"
            ],
    })


print(
    "\n=== SAFE LOCAL ALTERNATIVES ==="
)

print(
    "Count:",
    len(
        cell120_safe_alternatives
    )
)


for item in (
    cell120_safe_alternatives[
        :10
    ]
):

    print(
        item
    )


print(
    "\n=== INTERPRETATION ==="
)

print(
    "Safe alternative exists:",
    cell120_safe_alternative_exists
)

print(
    "Status:",
    cell120_status_detail
)

print(
    "Next action:",
    cell120_next_action
)

print(
    "Inspection SHA256:",
    cell120_local_window_audit_sha256
)


print(
    "\n=== FAILURES ==="
)

print(
    "Prerequisite failures:",
    cell120_prerequisite_failures
)

print(
    "Inspection failures:",
    cell120_failures
)


print(
    "\nSTATUS:",
    cell120_status
)


print(
    "\n=== CODE CELL 120 GATES ==="
)


for name, passed in (
    cell120_gates.items()
):

    print(
        name,
        ":",
        (
            "PASS"
            if passed
            else "FAIL"
        ),
    )


print(
    "\nFAILED GATES:",
    [
        name

        for name, passed
        in cell120_gates.items()

        if not passed
    ],
)


print(
    "\nCODE CELL 120 GATE:",
    (
        "PASS"
        if cell_120_pass
        else "FAIL"
    ),
)


assert cell_120_pass, (
    "Current-document local-window inspection failed. "
    "Do not retarget or advance the Queue. "
    "Repair this SAME Code Cell 120."
)
```

<h3>Cell 121 — بررسی Policy واقعی برای Pilot Case Replacement / Reselection</h3>

<p>
Cell 120 ثابت کرد در Document فعلی هیچ Local Window امنی برای Retarget وجود ندارد.
بنابراین مشکل دیگر با Retarget در همان سند حل نمی‌شود.
</p>

<p>
در این مرحله فقط بررسی می‌کنیم آیا در Contract، Runtime یا Notebook History یک مسیر واقعی برای
<code>Pilot Case Replacement</code>،
<code>Reselection</code>،
<code>Skip</code>
یا
<code>Reject-and-Advance</code>
تعریف شده است یا نه.
</p>

<p>
هیچ Pilot جایگزین نمی‌شود، Queue جلو نمی‌رود و هیچ State تغییر نمی‌کند.
</p>

</div>

```python
# Code Cell 121

# ============================================================
# Inspect existing policy / implementation for:
# - pilot replacement
# - pilot reselection
# - skip / reject-and-advance
#
# Read-only only.
# No queue advancement.
# No pilot replacement.
# No mutation.
# No artifact write.
# ============================================================


# ============================================================
# 1) Prerequisites
# ============================================================

cell121_prerequisite_failures = []


if globals().get("cell_120_pass") is not True:
    cell121_prerequisite_failures.append(
        "CELL120_NOT_PASS"
    )


if (
    globals().get("cell120_status")
    !=
    "CURRENT_DOCUMENT_LOCAL_WINDOWS_INSPECTED"
):
    cell121_prerequisite_failures.append({
        "reason": "CELL120_STATUS_UNEXPECTED",
        "actual": globals().get("cell120_status"),
    })


CELL121_REQUIRED_OBJECTS = [
    "cell120_local_window_audit",
    "PILOT_QA_WORKFLOW_CONTRACT_V1",
    "cell95_remaining_pilot_queue",
    "question_drafting_pilot_packets",
    "question_drafting_pilot_by_id",
    "cell97_question_draft_record",
]


cell121_missing_objects = [
    name
    for name in CELL121_REQUIRED_OBJECTS
    if name not in globals()
]


if cell121_missing_objects:
    cell121_prerequisite_failures.append({
        "reason": "REQUIRED_RUNTIME_OBJECTS_MISSING",
        "objects": cell121_missing_objects,
    })


# ============================================================
# 2) Helper
# ============================================================

def cell121_sha(value):
    return hashlib.sha256(
        stable_json_key(value).encode("utf-8")
    ).hexdigest()


# ============================================================
# 3) Freeze protected state BEFORE inspection
# ============================================================

cell121_before = {
    "queue": cell121_sha(
        cell95_remaining_pilot_queue
    ),
    "pilot": cell121_sha({
        "packets": question_drafting_pilot_packets,
        "by_id": question_drafting_pilot_by_id,
    }),
    "question": cell121_sha(
        cell97_question_draft_record
    ),
}


# ============================================================
# 4) Search workflow contract recursively
# ============================================================

CELL121_POLICY_TERMS = [
    "replace",
    "replacement",
    "reselect",
    "reselection",
    "skip",
    "reject",
    "advance",
    "blocked",
]


cell121_contract_hits = []


def cell121_walk(value, path=()):

    if isinstance(value, dict):

        for key, child in value.items():
            yield from cell121_walk(
                child,
                path + (str(key),),
            )

    elif isinstance(value, (list, tuple)):

        for index, child in enumerate(value):
            yield from cell121_walk(
                child,
                path + (str(index),),
            )

    else:
        yield path, value


for path, value in cell121_walk(
    PILOT_QA_WORKFLOW_CONTRACT_V1
):

    path_text = ".".join(path).lower()

    value_text = (
        value.lower()
        if isinstance(value, str)
        else ""
    )

    matched_terms = [
        term
        for term in CELL121_POLICY_TERMS
        if (
            term in path_text
            or term in value_text
        )
    ]

    if matched_terms:
        cell121_contract_hits.append({
            "path": ".".join(path),
            "value": value,
            "terms": matched_terms,
        })


# ============================================================
# 5) Read notebook history
# ============================================================

cell121_history_failures = []


try:
    cell121_history = list(
        get_ipython().history_manager.input_hist_raw
    )

except Exception as exc:
    cell121_history = []

    cell121_history_failures.append({
        "reason": "INPUT_HISTORY_READ_FAILED",
        "error": repr(exc),
    })


# ============================================================
# 6) Search prior cells for concrete replacement / reselection
#    implementation, not generic prose.
# ============================================================

CELL121_IMPLEMENTATION_TERMS = [
    "pilot_replacement",
    "replace_pilot",
    "replacement_pilot",
    "reselect_pilot",
    "pilot_reselection",
    "skip_pilot",
    "advance_queue",
    "remaining_pilot_queue",
]


cell121_history_hits = []


for history_index, source in enumerate(
    cell121_history
):

    if history_index >= 121:
        break

    if not isinstance(source, str):
        continue

    source_lower = source.lower()

    matched = [
        term
        for term in CELL121_IMPLEMENTATION_TERMS
        if term in source_lower
    ]

    if not matched:
        continue

    relevant_lines = [
        line.strip()
        for line in source.splitlines()
        if any(
            term in line.lower()
            for term in matched
        )
    ]

    cell121_history_hits.append({
        "history_index": history_index,
        "terms": matched,
        "relevant_lines": relevant_lines[:30],
    })


# ============================================================
# 7) Inspect Runtime object names only
# ============================================================

cell121_runtime_name_hits = []


for name, value in list(globals().items()):

    name_lower = str(name).lower()

    matched = [
        term
        for term in [
            "replace",
            "replacement",
            "reselect",
            "reselection",
            "skip_pilot",
            "advance_queue",
        ]
        if term in name_lower
    ]

    if matched:
        cell121_runtime_name_hits.append({
            "name": name,
            "type": type(value).__name__,
            "callable": callable(value),
            "terms": matched,
        })


cell121_runtime_name_hits = sorted(
    cell121_runtime_name_hits,
    key=lambda item: item["name"],
)


# ============================================================
# 8) Conservative classification
# ============================================================

cell121_has_explicit_contract_transition = any(
    any(
        term in hit["terms"]
        for term in [
            "replace",
            "replacement",
            "reselect",
            "reselection",
            "skip",
            "advance",
        ]
    )
    for hit in cell121_contract_hits
)


cell121_has_concrete_history_implementation = (
    len(
        cell121_history_hits
    )
    >
    0
)


cell121_has_runtime_primitive = any(
    item["callable"]
    for item in cell121_runtime_name_hits
)


if cell121_has_runtime_primitive:

    cell121_policy_status = (
        "PILOT_REPLACEMENT_RUNTIME_PRIMITIVE_CANDIDATE_FOUND"
    )

    cell121_next_action = (
        "INSPECT_EXACT_PILOT_REPLACEMENT_RUNTIME_PRIMITIVE"
    )


elif cell121_has_concrete_history_implementation:

    cell121_policy_status = (
        "PILOT_REPLACEMENT_HISTORY_IMPLEMENTATION_CANDIDATE_FOUND"
    )

    cell121_next_action = (
        "INSPECT_EXACT_PILOT_REPLACEMENT_HISTORY_IMPLEMENTATION"
    )


elif cell121_has_explicit_contract_transition:

    cell121_policy_status = (
        "PILOT_REPLACEMENT_CONTRACT_TRANSITION_CANDIDATE_FOUND"
    )

    cell121_next_action = (
        "INSPECT_EXACT_PILOT_REPLACEMENT_CONTRACT_TRANSITION"
    )


else:

    cell121_policy_status = (
        "NO_EXPLICIT_PILOT_REPLACEMENT_OR_RESELECTION_PATH_FOUND"
    )

    cell121_next_action = (
        "KEEP_CURRENT_PILOT_BLOCKED_DO_NOT_ADVANCE_QUEUE"
    )


# ============================================================
# 9) Result
# ============================================================

cell121_pilot_replacement_policy_inspection = {
    "inspection_version":
        "pilot_replacement_policy_inspection_v1",

    "pilot_case_id":
        cell97_question_draft_record.get(
            "pilot_case_id"
        ),

    "local_safe_alternative_exists":
        cell120_local_window_audit.get(
            "safe_alternative_exists"
        ),

    "contract_hit_count":
        len(
            cell121_contract_hits
        ),

    "history_hit_count":
        len(
            cell121_history_hits
        ),

    "runtime_name_hit_count":
        len(
            cell121_runtime_name_hits
        ),

    "has_explicit_contract_transition":
        cell121_has_explicit_contract_transition,

    "has_concrete_history_implementation":
        cell121_has_concrete_history_implementation,

    "has_runtime_primitive":
        cell121_has_runtime_primitive,

    "policy_status":
        cell121_policy_status,

    "pilot_replaced":
        False,

    "queue_advanced":
        False,

    "state_mutated":
        False,

    "next_action":
        cell121_next_action,
}


cell121_pilot_replacement_policy_sha256 = (
    cell121_sha(
        cell121_pilot_replacement_policy_inspection
    )
)


# ============================================================
# 10) Freeze protected state AFTER inspection
# ============================================================

cell121_after = {
    "queue": cell121_sha(
        cell95_remaining_pilot_queue
    ),
    "pilot": cell121_sha({
        "packets": question_drafting_pilot_packets,
        "by_id": question_drafting_pilot_by_id,
    }),
    "question": cell121_sha(
        cell97_question_draft_record
    ),
}


# ============================================================
# 11) Gates
# ============================================================

cell121_gates = {
    "cell120_is_pass":
        globals().get("cell_120_pass") is True,

    "required_objects_exist":
        len(cell121_missing_objects) == 0,

    "history_read_succeeded":
        len(cell121_history_failures) == 0,

    "contract_inspected":
        isinstance(
            cell121_contract_hits,
            list,
        ),

    "history_inspected":
        isinstance(
            cell121_history_hits,
            list,
        ),

    "runtime_names_inspected":
        isinstance(
            cell121_runtime_name_hits,
            list,
        ),

    "policy_classification_completed":
        isinstance(
            cell121_policy_status,
            str,
        ),

    "pilot_not_replaced":
        cell121_pilot_replacement_policy_inspection[
            "pilot_replaced"
        ]
        is False,

    "queue_not_advanced":
        cell121_pilot_replacement_policy_inspection[
            "queue_advanced"
        ]
        is False,

    "protected_state_byte_stable":
        cell121_before == cell121_after,

    "no_artifact_write":
        True,
}


cell_121_pass = all(
    cell121_gates.values()
)


cell121_status = (
    "PILOT_REPLACEMENT_POLICY_INSPECTED"
    if cell_121_pass
    else
    "PILOT_REPLACEMENT_POLICY_INSPECTION_FAILED"
)


# ============================================================
# 12) Compact diagnostics
# ============================================================

print(
    "=== PILOT REPLACEMENT / RESELECTION POLICY INSPECTION ==="
)


print(
    "\nCurrent Pilot:",
    cell97_question_draft_record.get(
        "pilot_case_id"
    )
)

print(
    "Safe local alternative exists:",
    cell120_local_window_audit.get(
        "safe_alternative_exists"
    )
)


print(
    "\n=== CONTRACT HITS ==="
)

print(
    "Count:",
    len(cell121_contract_hits)
)

for item in cell121_contract_hits[:20]:
    print(
        "-",
        item
    )


print(
    "\n=== HISTORY IMPLEMENTATION HITS ==="
)

print(
    "Count:",
    len(cell121_history_hits)
)

for item in cell121_history_hits[:10]:
    print(
        "\nHistory index:",
        item["history_index"]
    )
    print(
        "Terms:",
        item["terms"]
    )

    for line in item["relevant_lines"]:
        print(
            "-",
            line
        )


print(
    "\n=== RUNTIME NAME HITS ==="
)

print(
    "Count:",
    len(cell121_runtime_name_hits)
)

for item in cell121_runtime_name_hits[:20]:
    print(
        "-",
        item
    )


print(
    "\n=== INTERPRETATION ==="
)

print(
    "Explicit contract transition:",
    cell121_has_explicit_contract_transition
)

print(
    "Concrete history implementation:",
    cell121_has_concrete_history_implementation
)

print(
    "Runtime primitive:",
    cell121_has_runtime_primitive
)

print(
    "Policy status:",
    cell121_policy_status
)

print(
    "Next action:",
    cell121_next_action
)

print(
    "Inspection SHA256:",
    cell121_pilot_replacement_policy_sha256
)


print(
    "\n=== FAILURES ==="
)

print(
    "Prerequisite failures:",
    cell121_prerequisite_failures
)

print(
    "History failures:",
    cell121_history_failures
)


print(
    "\nSTATUS:",
    cell121_status
)


print(
    "\n=== CODE CELL 121 GATES ==="
)

for name, passed in cell121_gates.items():
    print(
        name,
        ":",
        "PASS"
        if passed
        else "FAIL",
    )


print(
    "\nFAILED GATES:",
    [
        name
        for name, passed
        in cell121_gates.items()
        if not passed
    ],
)


print(
    "\nCODE CELL 121 GATE:",
    "PASS"
    if cell_121_pass
    else "FAIL",
)


assert cell_121_pass, (
    "Pilot replacement/reselection policy inspection failed. "
    "Do not replace the Pilot or advance the Queue. "
    "Repair this SAME Code Cell 121."
)
```

```python
# Code Cell 124

# ============================================================
# Inspect construction of the CURRENT 23 Pilot packets and
# determine whether an eligible same-stratum replacement pool
# exists.
#
# Read-only only.
# No replacement selection.
# No pilot mutation.
# No queue advancement.
# No artifact write.
# ============================================================

import re
from collections import Counter


# ============================================================
# 1) Prerequisites
# ============================================================

cell124_prerequisite_failures = []


if globals().get("cell_123_pass") is not True:
    cell124_prerequisite_failures.append(
        "CELL123_NOT_PASS"
    )


if (
    globals().get("cell123_status")
    !=
    "PILOT_REPLACEMENT_REUSE_BOUNDARY_INSPECTED"
):
    cell124_prerequisite_failures.append({
        "reason": "CELL123_STATUS_UNEXPECTED",
        "actual": globals().get("cell123_status"),
    })


CELL124_REQUIRED_OBJECTS = [
    "question_drafting_pilot_packets",
    "question_drafting_pilot_by_id",
    "target_evidence_records",
    "golden_case_records",
    "task_draft_plan_rows",
    "cell95_remaining_pilot_queue",
    "cell97_question_draft_record",
]


cell124_missing_objects = [
    name
    for name in CELL124_REQUIRED_OBJECTS
    if name not in globals()
]


if cell124_missing_objects:
    cell124_prerequisite_failures.append({
        "reason": "REQUIRED_RUNTIME_OBJECTS_MISSING",
        "objects": cell124_missing_objects,
    })


# ============================================================
# 2) Helper
# ============================================================

def cell124_sha(value):
    return hashlib.sha256(
        stable_json_key(value).encode("utf-8")
    ).hexdigest()


# ============================================================
# 3) Freeze protected state BEFORE inspection
# ============================================================

cell124_before = {
    "pilots": cell124_sha(
        question_drafting_pilot_packets
    ),
    "pilot_index": cell124_sha(
        question_drafting_pilot_by_id
    ),
    "targets": cell124_sha(
        target_evidence_records
    ),
    "cases": cell124_sha(
        golden_case_records
    ),
    "tasks": cell124_sha(
        task_draft_plan_rows
    ),
    "queue": cell124_sha(
        cell95_remaining_pilot_queue
    ),
    "question": cell124_sha(
        cell97_question_draft_record
    ),
}


# ============================================================
# 4) Resolve current Pilot
# ============================================================

cell124_current_pilot_id = (
    cell97_question_draft_record.get(
        "pilot_case_id"
    )
)


cell124_current_packet = (
    question_drafting_pilot_by_id.get(
        cell124_current_pilot_id
    )
)


cell124_current_source_type = (
    cell124_current_packet.get(
        "source_type"
    )
    if isinstance(
        cell124_current_packet,
        dict,
    )
    else None
)


cell124_current_stratum = (
    cell124_current_packet.get(
        "stratum"
    )
    if isinstance(
        cell124_current_packet,
        dict,
    )
    else None
)


cell124_current_target_id = (
    cell124_current_packet.get(
        "target_evidence_id"
    )
    if isinstance(
        cell124_current_packet,
        dict,
    )
    else None
)


cell124_current_selection_basis = (
    cell124_current_packet.get(
        "target_selection_basis"
    )
    if isinstance(
        cell124_current_packet,
        dict,
    )
    else None
)


cell124_current_selection_method = (
    cell124_current_packet.get(
        "target_selection_method"
    )
    if isinstance(
        cell124_current_packet,
        dict,
    )
    else None
)


# ============================================================
# 5) Inspect current 23-Pilot population
# ============================================================

cell124_active_pilot_count = (
    len(
        question_drafting_pilot_packets
    )
)


cell124_selection_basis_counts = Counter(
    packet.get(
        "target_selection_basis"
    )
    for packet in question_drafting_pilot_packets
    if isinstance(
        packet,
        dict,
    )
)


cell124_selection_method_counts = Counter(
    packet.get(
        "target_selection_method"
    )
    for packet in question_drafting_pilot_packets
    if isinstance(
        packet,
        dict,
    )
)


cell124_pilot_stratum_counts = Counter(
    (
        packet.get(
            "source_type"
        ),
        packet.get(
            "stratum"
        ),
    )
    for packet in question_drafting_pilot_packets
    if isinstance(
        packet,
        dict,
    )
)


cell124_current_pilot_stratum_count = (
    cell124_pilot_stratum_counts.get(
        (
            cell124_current_source_type,
            cell124_current_stratum,
        ),
        0,
    )
)


# ============================================================
# 6) Build active Case / Task identity sets
# ============================================================

cell124_active_case_ids = {
    row.get(
        "golden_case_id"
    )
    for row in golden_case_records
    if (
        isinstance(
            row,
            dict,
        )
        and isinstance(
            row.get(
                "golden_case_id"
            ),
            str,
        )
    )
}


cell124_active_task_ids = {
    row.get(
        "task_draft_id"
    )
    for row in task_draft_plan_rows
    if (
        isinstance(
            row,
            dict,
        )
        and isinstance(
            row.get(
                "task_draft_id"
            ),
            str,
        )
    )
}


# ============================================================
# 7) Find same-source / same-stratum eligible Targets
#
# No ranking.
# No replacement selection.
# ============================================================

cell124_same_stratum_targets = []


for record in target_evidence_records:

    if not isinstance(
        record,
        dict,
    ):
        continue


    if (
        record.get(
            "source_type"
        )
        !=
        cell124_current_source_type
    ):
        continue


    if (
        record.get(
            "stratum"
        )
        !=
        cell124_current_stratum
    ):
        continue


    if (
        record.get(
            "golden_case_id"
        )
        not in
        cell124_active_case_ids
    ):
        continue


    if (
        record.get(
            "task_draft_id"
        )
        not in
        cell124_active_task_ids
    ):
        continue


    cell124_same_stratum_targets.append(
        record
    )


cell124_alternative_target_records = [
    record

    for record
    in cell124_same_stratum_targets

    if (
        record.get(
            "target_evidence_id"
        )
        !=
        cell124_current_target_id
    )
]


cell124_alternative_target_summary = [
    {
        "golden_case_id":
            record.get(
                "golden_case_id"
            ),

        "task_draft_id":
            record.get(
                "task_draft_id"
            ),

        "target_evidence_id":
            record.get(
                "target_evidence_id"
            ),

        "target_total_tokens":
            record.get(
                "target_total_tokens"
            ),

        "target_selection_basis":
            record.get(
                "target_selection_basis"
            ),
    }

    for record
    in cell124_alternative_target_records
]


# ============================================================
# 8) Read Notebook history
# ============================================================

cell124_history_failures = []


try:
    cell124_history = list(
        get_ipython().history_manager.input_hist_raw
    )

except Exception as exc:
    cell124_history = []

    cell124_history_failures.append({
        "reason": "INPUT_HISTORY_READ_FAILED",
        "error": repr(exc),
    })


# ============================================================
# 9) Locate actual Pilot construction cells
# ============================================================

CELL124_CONSTRUCTION_PATTERNS = [
    re.compile(
        r"\bquestion_drafting_pilot_packets\s*="
    ),
    re.compile(
        r"\bquestion_drafting_pilot_packets\.append\s*\("
    ),
    re.compile(
        r"\bquestion_drafting_pilot_by_id\s*="
    ),
]


cell124_construction_sources = []


for history_index, source in enumerate(
    cell124_history
):

    if not isinstance(
        source,
        str,
    ):
        continue


    if not any(
        pattern.search(
            source
        )
        for pattern in CELL124_CONSTRUCTION_PATTERNS
    ):
        continue


    relevant_lines = []


    for line in source.splitlines():

        lower = line.lower()


        if any(
            term in lower

            for term in [
                "question_drafting_pilot",
                "target_total_tokens",
                "target_selection_basis",
                "target_selection_method",
                "source_type",
                "stratum",
                "median",
                "distance",
                "sha256",
                "tie",
                "sorted",
                "pilot_case_id",
                "golden_case_id",
                "task_draft_id",
                "target_evidence_id",
            ]
        ):

            relevant_lines.append(
                line.strip()
            )


    cell124_construction_sources.append({
        "history_index":
            history_index,

        "relevant_lines":
            relevant_lines[
                :100
            ],

        "source":
            source,
    })


# ============================================================
# 10) Detect observable construction-rule components
# ============================================================

cell124_construction_text = "\n".join(
    item[
        "source"
    ]
    for item
    in cell124_construction_sources
)


CELL124_RULE_SIGNALS = [
    "target_total_tokens",
    "stratum",
    "target_evidence_id",
    "target_selection_basis",
    "sorted",
    "sha256",
    "median",
    "distance",
]


cell124_rule_signal_presence = {
    signal:
        (
            signal
            in
            cell124_construction_text
        )

    for signal
    in CELL124_RULE_SIGNALS
}


cell124_construction_found = (
    len(
        cell124_construction_sources
    )
    >
    0
)


cell124_has_core_selection_signals = all([
    cell124_rule_signal_presence[
        "target_total_tokens"
    ],
    cell124_rule_signal_presence[
        "stratum"
    ],
    cell124_rule_signal_presence[
        "target_evidence_id"
    ],
    cell124_rule_signal_presence[
        "target_selection_basis"
    ],
])


cell124_alternative_pool_exists = (
    len(
        cell124_alternative_target_records
    )
    >
    0
)


# ============================================================
# 11) Conservative interpretation
# ============================================================

if (
    cell124_construction_found
    and
    cell124_has_core_selection_signals
    and
    cell124_alternative_pool_exists
):

    cell124_construction_status = (
        "CURRENT_PILOT_SELECTION_CONSTRUCTION_AND_"
        "SAME_STRATUM_ALTERNATIVE_POOL_RESOLVED"
    )

    cell124_next_action = (
        "RECONSTRUCT_EXACT_CURRENT_PILOT_SELECTION_RULE_AND_"
        "DRY_RUN_WITH_BLOCKED_TARGET_EXCLUDED"
    )


elif (
    cell124_construction_found
    and
    not cell124_alternative_pool_exists
):

    cell124_construction_status = (
        "CURRENT_PILOT_SELECTION_CONSTRUCTION_FOUND_"
        "BUT_NO_SAME_STRATUM_ALTERNATIVE_POOL"
    )

    cell124_next_action = (
        "KEEP_CURRENT_PILOT_BLOCKED"
    )


else:

    cell124_construction_status = (
        "CURRENT_PILOT_SELECTION_CONSTRUCTION_NOT_FULLY_RESOLVED"
    )

    cell124_next_action = (
        "DO_NOT_REPLACE_OR_ADVANCE_QUEUE"
    )


# ============================================================
# 12) Result
# ============================================================

cell124_current_pilot_construction_inspection = {
    "inspection_version":
        "current_23_pilot_selection_construction_v1",

    "current_pilot_id":
        cell124_current_pilot_id,

    "active_pilot_count":
        cell124_active_pilot_count,

    "current_source_type":
        cell124_current_source_type,

    "current_stratum":
        cell124_current_stratum,

    "current_target_evidence_id":
        cell124_current_target_id,

    "current_selection_basis":
        cell124_current_selection_basis,

    "current_selection_method":
        cell124_current_selection_method,

    "current_pilot_stratum_count":
        cell124_current_pilot_stratum_count,

    "same_stratum_target_count":
        len(
            cell124_same_stratum_targets
        ),

    "alternative_target_count":
        len(
            cell124_alternative_target_records
        ),

    "construction_source_count":
        len(
            cell124_construction_sources
        ),

    "rule_signal_presence":
        cell124_rule_signal_presence,

    "construction_found":
        cell124_construction_found,

    "has_core_selection_signals":
        cell124_has_core_selection_signals,

    "alternative_pool_exists":
        cell124_alternative_pool_exists,

    "replacement_selected":
        False,

    "pilot_mutated":
        False,

    "queue_advanced":
        False,

    "status":
        cell124_construction_status,

    "next_action":
        cell124_next_action,
}


cell124_current_pilot_construction_sha256 = (
    cell124_sha(
        cell124_current_pilot_construction_inspection
    )
)


# ============================================================
# 13) Freeze protected state AFTER inspection
# ============================================================

cell124_after = {
    "pilots": cell124_sha(
        question_drafting_pilot_packets
    ),
    "pilot_index": cell124_sha(
        question_drafting_pilot_by_id
    ),
    "targets": cell124_sha(
        target_evidence_records
    ),
    "cases": cell124_sha(
        golden_case_records
    ),
    "tasks": cell124_sha(
        task_draft_plan_rows
    ),
    "queue": cell124_sha(
        cell95_remaining_pilot_queue
    ),
    "question": cell124_sha(
        cell97_question_draft_record
    ),
}


# ============================================================
# 14) Gates
# ============================================================

cell124_gates = {
    "cell123_is_pass":
        globals().get(
            "cell_123_pass"
        )
        is True,

    "required_objects_exist":
        len(
            cell124_missing_objects
        )
        ==
        0,

    "current_pilot_resolved":
        isinstance(
            cell124_current_packet,
            dict,
        ),

    "pilot_population_is_23":
        cell124_active_pilot_count
        ==
        23,

    "history_read_succeeded":
        len(
            cell124_history_failures
        )
        ==
        0,

    "construction_source_found":
        cell124_construction_found,

    "construction_rule_inspected":
        isinstance(
            cell124_rule_signal_presence,
            dict,
        ),

    "same_stratum_pool_inspected":
        isinstance(
            cell124_alternative_target_records,
            list,
        ),

    "alternative_pool_decision_completed":
        isinstance(
            cell124_alternative_pool_exists,
            bool,
        ),

    "replacement_not_selected":
        cell124_current_pilot_construction_inspection[
            "replacement_selected"
        ]
        is False,

    "pilot_not_mutated":
        cell124_current_pilot_construction_inspection[
            "pilot_mutated"
        ]
        is False,

    "queue_not_advanced":
        cell124_current_pilot_construction_inspection[
            "queue_advanced"
        ]
        is False,

    "protected_state_byte_stable":
        cell124_before
        ==
        cell124_after,

    "no_artifact_write":
        True,
}


cell_124_pass = all(
    cell124_gates.values()
)


cell124_status = (
    "CURRENT_23_PILOT_SELECTION_CONSTRUCTION_INSPECTED"

    if cell_124_pass

    else
    "CURRENT_23_PILOT_SELECTION_CONSTRUCTION_INSPECTION_FAILED"
)


# ============================================================
# 15) Compact diagnostics
# ============================================================

print(
    "=== CURRENT 23-PILOT SELECTION CONSTRUCTION ==="
)


print(
    "\n=== CURRENT PILOT ==="
)

print(
    "Pilot ID:",
    cell124_current_pilot_id
)

print(
    "Source type:",
    cell124_current_source_type
)

print(
    "Stratum:",
    cell124_current_stratum
)

print(
    "Target:",
    cell124_current_target_id
)

print(
    "Selection basis:",
    cell124_current_selection_basis
)

print(
    "Selection method:",
    cell124_current_selection_method
)


print(
    "\n=== ACTIVE PILOT POPULATION ==="
)

print(
    "Pilot count:",
    cell124_active_pilot_count
)

print(
    "Selection-basis counts:",
    dict(
        cell124_selection_basis_counts
    )
)

print(
    "Selection-method counts:",
    dict(
        cell124_selection_method_counts
    )
)

print(
    "Current stratum Pilot count:",
    cell124_current_pilot_stratum_count
)


print(
    "\n=== SAME-STRATUM TARGET POOL ==="
)

print(
    "Same-stratum Targets:",
    len(
        cell124_same_stratum_targets
    )
)

print(
    "Alternatives excluding current Target:",
    len(
        cell124_alternative_target_records
    )
)


for item in (
    cell124_alternative_target_summary[
        :10
    ]
):
    print(
        "-",
        item
    )


print(
    "\n=== PILOT CONSTRUCTION SOURCES ==="
)

print(
    "Count:",
    len(
        cell124_construction_sources
    )
)


for item in (
    cell124_construction_sources
):

    print(
        "\nHistory index:",
        item[
            "history_index"
        ]
    )

    for line in (
        item[
            "relevant_lines"
        ][
            :40
        ]
    ):
        print(
            "-",
            line
        )


print(
    "\n=== RULE SIGNALS ==="
)

for name, present in (
    cell124_rule_signal_presence.items()
):
    print(
        name,
        ":",
        present,
    )


print(
    "\n=== INTERPRETATION ==="
)

print(
    "Construction found:",
    cell124_construction_found
)

print(
    "Core selection signals present:",
    cell124_has_core_selection_signals
)

print(
    "Alternative pool exists:",
    cell124_alternative_pool_exists
)

print(
    "Status:",
    cell124_construction_status
)

print(
    "Next action:",
    cell124_next_action
)

print(
    "Inspection SHA256:",
    cell124_current_pilot_construction_sha256
)


print(
    "\n=== FAILURES ==="
)

print(
    "Prerequisite failures:",
    cell124_prerequisite_failures
)

print(
    "History failures:",
    cell124_history_failures
)


print(
    "\nSTATUS:",
    cell124_status
)


print(
    "\n=== CODE CELL 124 GATES ==="
)


for name, passed in (
    cell124_gates.items()
):
    print(
        name,
        ":",
        (
            "PASS"
            if passed
            else "FAIL"
        ),
    )


print(
    "\nFAILED GATES:",
    [
        name
        for name, passed
        in cell124_gates.items()
        if not passed
    ],
)


print(
    "\nCODE CELL 124 GATE:",
    (
        "PASS"
        if cell_124_pass
        else "FAIL"
    ),
)


assert cell_124_pass, (
    "Current 23-Pilot selection construction inspection failed. "
    "Do not replace the Pilot or advance the Queue. "
    "Repair this SAME Code Cell 124."
)
```

<h3>Cell 125 — بازتولید Rule انتخاب 23 Pilot و Dry Run با حذف Target بلاک‌شده</h3>

<p>
Cell 124 نشان داد برای Stratum فعلی 19 Target وجود دارد و با کنار گذاشتن Target فعلی،

این Winner صرفاً Candidate پیشنهادی Dry Run است و هیچ Replacement یا Activation انجام نمی‌شود.

```python
# Code Cell 125

# ============================================================
# Reproduce the exact 23-Pilot selection rule and perform
# a read-only exclusion Dry Run for the CURRENT blocked Target.
#
# No replacement staging.
# No activation.
# No queue advancement.
# No mutation.
# No artifact write.
# ============================================================

import inspect
from collections import defaultdict


# ============================================================
# 1) Prerequisites
# ============================================================

cell125_prerequisite_failures = []


if globals().get("cell_124_pass") is not True:
    cell125_prerequisite_failures.append(
        "CELL124_NOT_PASS"
    )


if (
    globals().get("cell124_status")
    !=
    "CURRENT_23_PILOT_SELECTION_CONSTRUCTION_INSPECTED"
):
    cell125_prerequisite_failures.append({
        "reason": "CELL124_STATUS_UNEXPECTED",
        "actual": globals().get("cell124_status"),
    })


CELL125_REQUIRED_OBJECTS = [
    "cell123_historical_source_rule",
    "cell124_history",
    "question_drafting_pilot_packets",
    "question_drafting_pilot_by_id",
    "target_evidence_records",
    "cell97_question_draft_record",
    "cell95_remaining_pilot_queue",
    "stable_json_key",
]


cell125_missing_objects = [
    name
    for name in CELL125_REQUIRED_OBJECTS
    if name not in globals()
]


if cell125_missing_objects:
    cell125_prerequisite_failures.append({
        "reason": "REQUIRED_RUNTIME_OBJECTS_MISSING",
        "objects": cell125_missing_objects,
    })


# ============================================================
# 2) Helper
# ============================================================

def cell125_sha(value):
    return hashlib.sha256(
        stable_json_key(value).encode("utf-8")
    ).hexdigest()


# ============================================================
# 3) Freeze protected state
# ============================================================

cell125_before = {
    "pilots": cell125_sha(
        question_drafting_pilot_packets
    ),
    "pilot_index": cell125_sha(
        question_drafting_pilot_by_id
    ),
    "targets": cell125_sha(
        target_evidence_records
    ),
    "queue": cell125_sha(
        cell95_remaining_pilot_queue
    ),
    "question": cell125_sha(
        cell97_question_draft_record
    ),
}


# ============================================================
# 4) Resolve exact current construction source
# ============================================================

cell125_rule_failures = []


cell125_history29_source = (
    cell124_history[29]
    if (
        isinstance(cell124_history, list)
        and len(cell124_history) > 29
        and isinstance(cell124_history[29], str)
    )
    else None
)


if cell125_history29_source is None:
    cell125_rule_failures.append(
        "HISTORY_29_SOURCE_NOT_RESOLVED"
    )


# ============================================================
# 5) Resolve nearest-rank callable
# ============================================================

cell125_nearest_rank_name = None
cell125_nearest_rank_callable = None


if callable(
    globals().get(
        "empirical_nearest_rank"
    )
):
    cell125_nearest_rank_name = (
        "empirical_nearest_rank"
    )
    cell125_nearest_rank_callable = (
        globals()[
            "empirical_nearest_rank"
        ]
    )

else:
    cell125_nearest_rank_candidates = [
        name
        for name, value in globals().items()
        if (
            callable(value)
            and "nearest" in str(name).lower()
            and "rank" in str(name).lower()
        )
    ]

    if len(cell125_nearest_rank_candidates) == 1:
        cell125_nearest_rank_name = (
            cell125_nearest_rank_candidates[0]
        )
        cell125_nearest_rank_callable = globals()[
            cell125_nearest_rank_name
        ]


if cell125_nearest_rank_callable is None:
    cell125_rule_failures.append(
        "EMPIRICAL_NEAREST_RANK_CALLABLE_NOT_RESOLVED"
    )


# ============================================================
# 6) Confirm serializer from actual Cell 29 source
# ============================================================

cell125_stable_serializer_confirmed = (
    isinstance(
        cell125_history29_source,
        str,
    )
    and
    "stable_json_key" in cell125_history29_source
)


if not cell125_stable_serializer_confirmed:
    cell125_rule_failures.append(
        "CELL29_STABLE_JSON_SERIALIZER_NOT_CONFIRMED"
    )


# ============================================================
# 7) Confirm rule components from actual Cell 29 source
# ============================================================

CELL125_REQUIRED_RULE_SIGNALS = [
    "stratum_median_tokens",
    "distance_to_median",
    "stable_tie_break",
    "target_total_tokens",
    "target_evidence_id",
    "sorted",
]


cell125_rule_signal_presence = {
    signal: (
        isinstance(
            cell125_history29_source,
            str,
        )
        and signal in cell125_history29_source
    )
    for signal in CELL125_REQUIRED_RULE_SIGNALS
}


if not all(
    cell125_rule_signal_presence.values()
):
    cell125_rule_failures.append({
        "reason": "CELL29_RULE_SIGNALS_INCOMPLETE",
        "signals": cell125_rule_signal_presence,
    })


# ============================================================
# 8) Build Target pools only for the 23 active Pilot strata
# ============================================================

cell125_pilot_by_stratum = {}


for packet in question_drafting_pilot_packets:

    if not isinstance(packet, dict):
        continue

    key = (
        packet.get("source_type"),
        packet.get("stratum"),
    )

    cell125_pilot_by_stratum[
        key
    ] = packet


cell125_active_strata = sorted(
    cell125_pilot_by_stratum.keys(),
    key=lambda item: (
        str(item[0]),
        str(item[1]),
    ),
)


cell125_targets_by_stratum = defaultdict(
    list
)


for record in target_evidence_records:

    if not isinstance(record, dict):
        continue

    key = (
        record.get("source_type"),
        record.get("stratum"),
    )

    if key in cell125_pilot_by_stratum:
        cell125_targets_by_stratum[
            key
        ].append(
            record
        )


# ============================================================
# 9) Generic selection-rule function
# ============================================================

def cell125_select_winner(
    source_type,
    stratum,
    records,
):

    assert records


    token_values = [
        record[
            "target_total_tokens"
        ]
        for record in records
    ]


    stratum_median_tokens = (
        cell125_nearest_rank_callable(
            token_values,
            0.50,
        )
    )


    ranking_rows = []


    for record in records:

        target_evidence_id = (
            record[
                "target_evidence_id"
            ]
        )


        distance_to_median = abs(
            record[
                "target_total_tokens"
            ]
            -
            stratum_median_tokens
        )


        stable_tie_break = hashlib.sha256(
            stable_json_key({
                "source_type":
                    source_type,

                "stratum":
                    stratum,

                "target_evidence_id":
                    target_evidence_id,
            }).encode(
                "utf-8"
            )
        ).hexdigest()


        ranking_rows.append({
            "record":
                record,

            "distance_to_median":
                distance_to_median,

            "stable_tie_break":
                stable_tie_break,
        })


    winner = sorted(
        ranking_rows,
        key=lambda item: (
            item[
                "distance_to_median"
            ],
            item[
                "stable_tie_break"
            ],
        ),
    )[0]


    return {
        "median":
            stratum_median_tokens,

        "winner":
            winner,

        "ranking":
            ranking_rows,
    }


# ============================================================
# 10) Reproduce all 23 active Pilot selections
# ============================================================

cell125_reproduction_rows = []
cell125_reproduction_failures = []


if len(cell125_rule_failures) == 0:

    for key in cell125_active_strata:

        source_type, stratum = key


        pool = (
            cell125_targets_by_stratum.get(
                key,
                [],
            )
        )


        persisted_packet = (
            cell125_pilot_by_stratum[
                key
            ]
        )


        if not pool:
            cell125_reproduction_failures.append({
                "source_type":
                    source_type,

                "stratum":
                    stratum,

                "reason":
                    "EMPTY_TARGET_POOL",
            })

            continue


        try:

            reproduction = (
                cell125_select_winner(
                    source_type,
                    stratum,
                    pool,
                )
            )


        except Exception as exc:

            cell125_reproduction_failures.append({
                "source_type":
                    source_type,

                "stratum":
                    stratum,

                "reason":
                    "RULE_EXECUTION_FAILED",

                "error":
                    repr(
                        exc
                    ),
            })

            continue


        reproduced_id = (
            reproduction[
                "winner"
            ][
                "record"
            ][
                "target_evidence_id"
            ]
        )


        persisted_id = (
            persisted_packet[
                "target_evidence_id"
            ]
        )


        cell125_reproduction_rows.append({
            "source_type":
                source_type,

            "stratum":
                stratum,

            "pool_count":
                len(
                    pool
                ),

            "median":
                reproduction[
                    "median"
                ],

            "persisted_target_evidence_id":
                persisted_id,

            "reproduced_target_evidence_id":
                reproduced_id,

            "exact_match":
                persisted_id
                ==
                reproduced_id,
        })


cell125_exact_reproduction_count = sum(
    1
    for row in cell125_reproduction_rows
    if row[
        "exact_match"
    ]
)


cell125_all_23_reproduced = (
    len(
        cell125_reproduction_rows
    )
    ==
    23

    and

    cell125_exact_reproduction_count
    ==
    23

    and

    len(
        cell125_reproduction_failures
    )
    ==
    0
)


# ============================================================
# 11) Current blocked Pilot / Stratum
# ============================================================

cell125_current_pilot_id = (
    cell97_question_draft_record.get(
        "pilot_case_id"
    )
)


cell125_current_packet = (
    question_drafting_pilot_by_id[
        cell125_current_pilot_id
    ]
)


cell125_current_source_type = (
    cell125_current_packet[
        "source_type"
    ]
)


cell125_current_stratum = (
    cell125_current_packet[
        "stratum"
    ]
)


cell125_blocked_target_id = (
    cell125_current_packet[
        "target_evidence_id"
    ]
)


cell125_current_key = (
    cell125_current_source_type,
    cell125_current_stratum,
)


cell125_current_pool = list(
    cell125_targets_by_stratum[
        cell125_current_key
    ]
)


# ============================================================
# 12) Exclusion Dry Run
# ============================================================

cell125_exclusion_pool = [
    record

    for record
    in cell125_current_pool

    if (
        record.get(
            "target_evidence_id"
        )
        !=
        cell125_blocked_target_id
    )
]


cell125_exclusion_result = None
cell125_exclusion_failure = None


if (
    cell125_all_23_reproduced
    and
    len(
        cell125_exclusion_pool
    )
    >
    0
):

    try:

        cell125_exclusion_result = (
            cell125_select_winner(
                cell125_current_source_type,
                cell125_current_stratum,
                cell125_exclusion_pool,
            )
        )


    except Exception as exc:

        cell125_exclusion_failure = {
            "reason":
                "EXCLUSION_DRY_RUN_FAILED",

            "error":
                repr(
                    exc
                ),
        }


# ============================================================
# 13) Resolve hypothetical next Candidate
# ============================================================

cell125_exclusion_winner = (
    cell125_exclusion_result[
        "winner"
    ]

    if isinstance(
        cell125_exclusion_result,
        dict,
    )

    else None
)


cell125_exclusion_candidate_record = (
    cell125_exclusion_winner[
        "record"
    ]

    if isinstance(
        cell125_exclusion_winner,
        dict,
    )

    else None
)


cell125_exclusion_candidate_summary = (
    {
        "golden_case_id":
            cell125_exclusion_candidate_record.get(
                "golden_case_id"
            ),

        "task_draft_id":
            cell125_exclusion_candidate_record.get(
                "task_draft_id"
            ),

        "target_evidence_id":
            cell125_exclusion_candidate_record.get(
                "target_evidence_id"
            ),

        "target_total_tokens":
            cell125_exclusion_candidate_record.get(
                "target_total_tokens"
            ),

        "target_selection_basis":
            cell125_exclusion_candidate_record.get(
                "target_selection_basis"
            ),

        "distance_to_median":
            cell125_exclusion_winner.get(
                "distance_to_median"
            ),

        "stable_tie_break":
            cell125_exclusion_winner.get(
                "stable_tie_break"
            ),
    }

    if isinstance(
        cell125_exclusion_candidate_record,
        dict,
    )

    else None
)


cell125_exclusion_candidate_exists = (
    isinstance(
        cell125_exclusion_candidate_summary,
        dict,
    )

    and

    cell125_exclusion_candidate_summary.get(
        "target_evidence_id"
    )
    !=
    cell125_blocked_target_id
)


# ============================================================
# 14) Ensure Candidate is not already another active Pilot
# ============================================================

cell125_active_pilot_target_ids = {
    packet.get(
        "target_evidence_id"
    )

    for packet
    in question_drafting_pilot_packets

    if isinstance(
        packet,
        dict,
    )
}


cell125_candidate_already_active_elsewhere = (
    cell125_exclusion_candidate_exists

    and

    cell125_exclusion_candidate_summary[
        "target_evidence_id"
    ]
    in
    (
        cell125_active_pilot_target_ids
        -
        {
            cell125_blocked_target_id
        }
    )
)


# ============================================================
# 15) Interpretation
# ============================================================

if (
    cell125_all_23_reproduced
    and
    cell125_exclusion_candidate_exists
    and
    not cell125_candidate_already_active_elsewhere
):

    cell125_status_detail = (
        "CURRENT_PILOT_RULE_EXACTLY_REPRODUCED_"
        "AND_EXCLUSION_CANDIDATE_RESOLVED"
    )

    cell125_next_action = (
        "INSPECT_EXCLUSION_WINNER_FULL_CASE_TASK_TARGET_"
        "AND_GROUNDABILITY_BEFORE_ANY_REPLACEMENT_STAGING"
    )


elif not cell125_all_23_reproduced:

    cell125_status_detail = (
        "CURRENT_PILOT_SELECTION_RULE_NOT_EXACTLY_REPRODUCED"
    )

    cell125_next_action = (
        "DO_NOT_USE_EXCLUSION_CANDIDATE"
    )


else:

    cell125_status_detail = (
        "EXCLUSION_DRY_RUN_DID_NOT_RESOLVE_SAFE_UNIQUE_CANDIDATE"
    )

    cell125_next_action = (
        "KEEP_CURRENT_PILOT_BLOCKED"
    )


# ============================================================
# 16) Result
# ============================================================

cell125_exclusion_dry_run = {
    "inspection_version":
        "current_pilot_exclusion_dry_run_v1",

    "nearest_rank_callable":
        cell125_nearest_rank_name,

    "stable_serializer_confirmed":
        cell125_stable_serializer_confirmed,

    "active_stratum_count":
        len(
            cell125_active_strata
        ),

    "reproduction_row_count":
        len(
            cell125_reproduction_rows
        ),

    "exact_reproduction_count":
        cell125_exact_reproduction_count,

    "all_23_reproduced":
        cell125_all_23_reproduced,

    "current_pilot_id":
        cell125_current_pilot_id,

    "blocked_target_evidence_id":
        cell125_blocked_target_id,

    "current_pool_count":
        len(
            cell125_current_pool
        ),

    "exclusion_pool_count":
        len(
            cell125_exclusion_pool
        ),

    "exclusion_candidate":
        cell125_exclusion_candidate_summary,

    "candidate_already_active_elsewhere":
        cell125_candidate_already_active_elsewhere,

    "replacement_staged":
        False,

    "replacement_activated":
        False,

    "queue_advanced":
        False,

    "status":
        cell125_status_detail,

    "next_action":
        cell125_next_action,
}


cell125_exclusion_dry_run_sha256 = (
    cell125_sha(
        cell125_exclusion_dry_run
    )
)


# ============================================================
# 17) Freeze protected state AFTER inspection
# ============================================================

cell125_after = {
    "pilots": cell125_sha(
        question_drafting_pilot_packets
    ),
    "pilot_index": cell125_sha(
        question_drafting_pilot_by_id
    ),
    "targets": cell125_sha(
        target_evidence_records
    ),
    "queue": cell125_sha(
        cell95_remaining_pilot_queue
    ),
    "question": cell125_sha(
        cell97_question_draft_record
    ),
}


# ============================================================
# 18) Gates
# ============================================================

cell125_gates = {
    "cell124_is_pass":
        globals().get(
            "cell_124_pass"
        )
        is True,

    "required_objects_exist":
        len(
            cell125_missing_objects
        )
        ==
        0,

    "cell29_source_resolved":
        isinstance(
            cell125_history29_source,
            str,
        ),

    "nearest_rank_callable_resolved":
        callable(
            cell125_nearest_rank_callable
        ),

    "stable_serializer_confirmed":
        cell125_stable_serializer_confirmed,

    "rule_signals_confirmed":
        all(
            cell125_rule_signal_presence.values()
        ),

    "all_23_active_pilots_reproduced":
        cell125_all_23_reproduced,

    "exclusion_pool_nonempty":
        len(
            cell125_exclusion_pool
        )
        >
        0,

    "exclusion_dry_run_succeeded":
        (
            cell125_exclusion_failure
            is None
        ),

    "exclusion_candidate_resolved":
        cell125_exclusion_candidate_exists,

    "candidate_not_already_active_elsewhere":
        not cell125_candidate_already_active_elsewhere,

    "replacement_not_staged":
        cell125_exclusion_dry_run[
            "replacement_staged"
        ]
        is False,

    "replacement_not_activated":
        cell125_exclusion_dry_run[
            "replacement_activated"
        ]
        is False,

    "queue_not_advanced":
        cell125_exclusion_dry_run[
            "queue_advanced"
        ]
        is False,

    "protected_state_byte_stable":
        cell125_before
        ==
        cell125_after,

    "no_artifact_write":
        True,
}


cell_125_pass = all(
    cell125_gates.values()
)


cell125_status = (
    "CURRENT_PILOT_SELECTION_RULE_AND_EXCLUSION_DRY_RUN_INSPECTED"

    if cell_125_pass

    else
    "CURRENT_PILOT_SELECTION_RULE_AND_EXCLUSION_DRY_RUN_FAILED"
)


# ============================================================
# 19) Diagnostics
# ============================================================

print(
    "=== CURRENT PILOT RULE REPRODUCTION + EXCLUSION DRY RUN ==="
)


print(
    "\n=== RULE RESOLUTION ==="
)

print(
    "Nearest-rank callable:",
    cell125_nearest_rank_name
)

print(
    "Stable serializer confirmed:",
    cell125_stable_serializer_confirmed
)

print(
    "Rule signals:",
    cell125_rule_signal_presence
)


print(
    "\n=== 23-PILOT REPRODUCTION ==="
)

print(
    "Active strata:",
    len(
        cell125_active_strata
    )
)

print(
    "Reproduction rows:",
    len(
        cell125_reproduction_rows
    )
)

print(
    "Exact matches:",
    cell125_exact_reproduction_count
)

print(
    "All 23 reproduced:",
    cell125_all_23_reproduced
)


for row in cell125_reproduction_rows:

    if not row[
        "exact_match"
    ]:

        print(
            "MISMATCH:",
            row
        )


print(
    "\nReproduction failures:",
    cell125_reproduction_failures
)


print(
    "\n=== CURRENT STRATUM EXCLUSION ==="
)

print(
    "Source type:",
    cell125_current_source_type
)

print(
    "Stratum:",
    cell125_current_stratum
)

print(
    "Blocked Target:",
    cell125_blocked_target_id
)

print(
    "Original pool count:",
    len(
        cell125_current_pool
    )
)

print(
    "Exclusion pool count:",
    len(
        cell125_exclusion_pool
    )
)

print(
    "Exclusion median:",
    (
        cell125_exclusion_result.get(
            "median"
        )

        if isinstance(
            cell125_exclusion_result,
            dict,
        )

        else None
    )
)


print(
    "\n=== DRY-RUN NEXT CANDIDATE ==="
)

print(
    cell125_exclusion_candidate_summary
)

print(
    "Already active elsewhere:",
    cell125_candidate_already_active_elsewhere
)


print(
    "\n=== INTERPRETATION ==="
)

print(
    "Status:",
    cell125_status_detail
)

print(
    "Next action:",
    cell125_next_action
)

print(
    "Inspection SHA256:",
    cell125_exclusion_dry_run_sha256
)


print(
    "\n=== FAILURES ==="
)

print(
    "Prerequisite failures:",
    cell125_prerequisite_failures
)

print(
    "Rule failures:",
    cell125_rule_failures
)

print(
    "Exclusion failure:",
    cell125_exclusion_failure
)


print(
    "\nSTATUS:",
    cell125_status
)


print(
    "\n=== CODE CELL 125 GATES ==="
)


for name, passed in (
    cell125_gates.items()
):

    print(
        name,
        ":",
        (
            "PASS"
            if passed
            else "FAIL"
        ),
    )


print(
    "\nFAILED GATES:",
    [
        name
        for name, passed
        in cell125_gates.items()
        if not passed
    ],
)


print(
    "\nCODE CELL 125 GATE:",
    (
        "PASS"
        if cell_125_pass
        else "FAIL"
    ),
)


assert cell_125_pass, (
    "Pilot selection-rule reproduction or exclusion Dry Run failed. "
    "Do not stage a replacement or advance the Queue. "
    "Repair this SAME Code Cell 125."
)
```

<h3>Cell 126 — بررسی کامل Case / Task / Target / Groundability برای Candidate جایگزین</h3>

<p>
Cell 125 Rule واقعی انتخاب 23 Pilot را دقیقاً برای هر 23 مورد بازتولید کرد و پس از حذف Target بلاک‌شده،
یک Candidate جایگزین مشخص شد.
</p>

<p>
در این مرحله هیچ Replacement یا Staging انجام نمی‌شود.
فقط بررسی می‌کنیم Golden Case، Task Draft و Target Evidence مربوط به Candidate همگی دقیقاً یک رکورد داشته باشند،
شناسه‌هایشان با هم سازگار باشد و Source Type و Stratum آن‌ها همان Stratum مورد انتظار باشد.
</p>

<p>
همچنین Runtime Objectهای موجود که به Groundability مربوط‌اند به‌صورت Read-Only بررسی می‌شوند تا ببینیم
برای همین Candidate چه Groundability Record یا Status واقعی وجود دارد.
</p>

<p>
هیچ Pilot، Queue، Question، Target یا Artifact تغییر نمی‌کند.
</p>

```python
# Code Cell 126

# ============================================================
# Inspect the exclusion winner's full:
#   Case -> Task -> Target -> Groundability chain
#
# Read-only only.
# No replacement staging.
# No activation.
# No queue advancement.
# No mutation.
# No artifact write.
# ============================================================


# ============================================================
# 1) Prerequisites
# ============================================================

cell126_prerequisite_failures = []


if globals().get(
    "cell_125_pass"
) is not True:

    cell126_prerequisite_failures.append(
        "CELL125_NOT_PASS"
    )


if (
    globals().get(
        "cell125_status"
    )
    !=
    "CURRENT_PILOT_SELECTION_RULE_AND_EXCLUSION_DRY_RUN_INSPECTED"
):

    cell126_prerequisite_failures.append({
        "reason":
            "CELL125_STATUS_UNEXPECTED",

        "actual":
            globals().get(
                "cell125_status"
            ),
    })


CELL126_REQUIRED_OBJECTS = [
    "cell125_exclusion_candidate_summary",
    "cell125_current_source_type",
    "cell125_current_stratum",
    "golden_case_records",
    "task_draft_plan_rows",
    "target_evidence_records",
    "question_drafting_pilot_packets",
    "question_drafting_pilot_by_id",
    "cell95_remaining_pilot_queue",
    "cell97_question_draft_record",
]


cell126_missing_objects = [
    name

    for name
    in CELL126_REQUIRED_OBJECTS

    if name not in globals()
]


if cell126_missing_objects:

    cell126_prerequisite_failures.append({
        "reason":
            "REQUIRED_RUNTIME_OBJECTS_MISSING",

        "objects":
            cell126_missing_objects,
    })


# ============================================================
# 2) Helper
# ============================================================

def cell126_sha(value):

    return hashlib.sha256(
        stable_json_key(
            value
        ).encode(
            "utf-8"
        )
    ).hexdigest()


def cell126_status_fields(
    record,
):

    if not isinstance(
        record,
        dict,
    ):

        return {}


    return {
        key:
            value

        for key, value
        in record.items()

        if (
            isinstance(
                key,
                str,
            )

            and

            (
                "status"
                in
                key.lower()

                or

                "ground"
                in
                key.lower()

                or

                "valid"
                in
                key.lower()
            )
        )
    }


# ============================================================
# 3) Freeze protected state BEFORE inspection
# ============================================================

cell126_before = {
    "pilots":
        cell126_sha(
            question_drafting_pilot_packets
        ),

    "pilot_index":
        cell126_sha(
            question_drafting_pilot_by_id
        ),

    "cases":
        cell126_sha(
            golden_case_records
        ),

    "tasks":
        cell126_sha(
            task_draft_plan_rows
        ),

    "targets":
        cell126_sha(
            target_evidence_records
        ),

    "queue":
        cell126_sha(
            cell95_remaining_pilot_queue
        ),

    "question":
        cell126_sha(
            cell97_question_draft_record
        ),
}


# ============================================================
# 4) Resolve Candidate identity
# ============================================================

cell126_candidate = (
    cell125_exclusion_candidate_summary
)


cell126_candidate_case_id = (
    cell126_candidate.get(
        "golden_case_id"
    )

    if isinstance(
        cell126_candidate,
        dict,
    )

    else None
)


cell126_candidate_task_id = (
    cell126_candidate.get(
        "task_draft_id"
    )

    if isinstance(
        cell126_candidate,
        dict,
    )

    else None
)


cell126_candidate_target_id = (
    cell126_candidate.get(
        "target_evidence_id"
    )

    if isinstance(
        cell126_candidate,
        dict,
    )

    else None
)


# ============================================================
# 5) Resolve exact Case / Task / Target records
# ============================================================

cell126_case_matches = [
    record

    for record
    in golden_case_records

    if (
        isinstance(
            record,
            dict,
        )

        and

        record.get(
            "golden_case_id"
        )
        ==
        cell126_candidate_case_id
    )
]


cell126_task_matches = [
    record

    for record
    in task_draft_plan_rows

    if (
        isinstance(
            record,
            dict,
        )

        and

        record.get(
            "task_draft_id"
        )
        ==
        cell126_candidate_task_id
    )
]


cell126_target_matches = [
    record

    for record
    in target_evidence_records

    if (
        isinstance(
            record,
            dict,
        )

        and

        record.get(
            "target_evidence_id"
        )
        ==
        cell126_candidate_target_id
    )
]


cell126_case_record = (
    cell126_case_matches[
        0
    ]

    if len(
        cell126_case_matches
    )
    ==
    1

    else None
)


cell126_task_record = (
    cell126_task_matches[
        0
    ]

    if len(
        cell126_task_matches
    )
    ==
    1

    else None
)


cell126_target_record = (
    cell126_target_matches[
        0
    ]

    if len(
        cell126_target_matches
    )
    ==
    1

    else None
)


# ============================================================
# 6) Identity-chain checks
# ============================================================

cell126_case_task_link = (
    isinstance(
        cell126_task_record,
        dict,
    )

    and

    cell126_task_record.get(
        "golden_case_id"
    )
    ==
    cell126_candidate_case_id
)


cell126_target_case_link = (
    isinstance(
        cell126_target_record,
        dict,
    )

    and

    cell126_target_record.get(
        "golden_case_id"
    )
    ==
    cell126_candidate_case_id
)


cell126_target_task_link = (
    isinstance(
        cell126_target_record,
        dict,
    )

    and

    cell126_target_record.get(
        "task_draft_id"
    )
    ==
    cell126_candidate_task_id
)


cell126_source_type_match = (
    isinstance(
        cell126_target_record,
        dict,
    )

    and

    cell126_target_record.get(
        "source_type"
    )
    ==
    cell125_current_source_type
)


cell126_stratum_match = (
    isinstance(
        cell126_target_record,
        dict,
    )

    and

    cell126_target_record.get(
        "stratum"
    )
    ==
    cell125_current_stratum
)


# ============================================================
# 7) Target structural checks
# ============================================================

cell126_target_parts = (
    cell126_target_record.get(
        "target_parts"
    )

    if isinstance(
        cell126_target_record,
        dict,
    )

    else None
)


cell126_target_parts_present = (
    isinstance(
        cell126_target_parts,
        list,
    )

    and

    len(
        cell126_target_parts
    )
    >
    0
)


cell126_target_token_count = (
    cell126_target_record.get(
        "target_total_tokens"
    )

    if isinstance(
        cell126_target_record,
        dict,
    )

    else None
)


cell126_target_token_match = (
    cell126_target_token_count
    ==
    cell126_candidate.get(
        "target_total_tokens"
    )
)


cell126_target_selection_basis_match = (
    isinstance(
        cell126_target_record,
        dict,
    )

    and

    cell126_target_record.get(
        "target_selection_basis"
    )
    ==
    cell126_candidate.get(
        "target_selection_basis"
    )
)


# ============================================================
# 8) Compact status views
# ============================================================

cell126_case_status_view = (
    cell126_status_fields(
        cell126_case_record
    )
)


cell126_task_status_view = (
    cell126_status_fields(
        cell126_task_record
    )
)


cell126_target_status_view = (
    cell126_status_fields(
        cell126_target_record
    )
)


# ============================================================
# 9) Discover Groundability Runtime objects
#
# Only shallow inspection of globals whose NAME contains
# "groundability".
# ============================================================

cell126_groundability_runtime_objects = []


for name, value in list(
    globals().items()
):

    if (
        "groundability"
        not in
        str(
            name
        ).lower()
    ):

        continue


    if name.startswith(
        "cell126_"
    ):

        continue


    cell126_groundability_runtime_objects.append({
        "name":
            name,

        "type":
            type(
                value
            ).__name__,

        "size":
            (
                len(
                    value
                )

                if isinstance(
                    value,
                    (
                        list,
                        tuple,
                        dict,
                    ),
                )

                else None
            ),
    })


cell126_groundability_runtime_objects = sorted(
    cell126_groundability_runtime_objects,
    key=lambda item: item[
        "name"
    ],
)


# ============================================================
# 10) Find exact Groundability records for Candidate
# ============================================================

cell126_groundability_matches = []


def cell126_record_matches_candidate(
    record,
):

    if not isinstance(
        record,
        dict,
    ):

        return False


    identifiers = {
        "golden_case_id":
            cell126_candidate_case_id,

        "task_draft_id":
            cell126_candidate_task_id,

        "target_evidence_id":
            cell126_candidate_target_id,
    }


    present_id_fields = [
        key

        for key
        in identifiers

        if key in record
    ]


    if not present_id_fields:

        return False


    return all(
        record.get(
            key
        )
        ==
        identifiers[
            key
        ]

        for key
        in present_id_fields
    )


for object_info in (
    cell126_groundability_runtime_objects
):

    name = (
        object_info[
            "name"
        ]
    )


    value = globals()[
        name
    ]


    candidate_records = []


    if isinstance(
        value,
        list,
    ):

        candidate_records = value


    elif isinstance(
        value,
        tuple,
    ):

        candidate_records = list(
            value
        )


    elif isinstance(
        value,
        dict,
    ):

        # Include the dict itself if it is a record.
        candidate_records = [
            value
        ]


        # Also inspect direct dict values only.
        candidate_records.extend(
            child

            for child
            in value.values()

            if isinstance(
                child,
                dict,
            )
        )


    matches = [
        record

        for record
        in candidate_records

        if cell126_record_matches_candidate(
            record
        )
    ]


    for record in matches:

        cell126_groundability_matches.append({
            "runtime_object":
                name,

            "record":
                record,

            "status_fields":
                cell126_status_fields(
                    record
                ),
        })


# ============================================================
# 11) Groundability interpretation
# ============================================================

cell126_groundability_match_count = (
    len(
        cell126_groundability_matches
    )
)


cell126_groundability_status_values = []


for match in (
    cell126_groundability_matches
):

    for key, value in (
        match[
            "status_fields"
        ].items()
    ):

        cell126_groundability_status_values.append({
            "runtime_object":
                match[
                    "runtime_object"
                ],

            "field":
                key,

            "value":
                value,
        })


cell126_groundability_evidence_found = (
    cell126_groundability_match_count
    >
    0
)


# ============================================================
# 12) Check whether Candidate is already an active Pilot
# ============================================================

cell126_active_pilot_target_ids = {
    packet.get(
        "target_evidence_id"
    )

    for packet
    in question_drafting_pilot_packets

    if isinstance(
        packet,
        dict,
    )
}


cell126_candidate_is_active_pilot = (
    cell126_candidate_target_id
    in
    cell126_active_pilot_target_ids
)


# ============================================================
# 13) Conservative chain decision
# ============================================================

cell126_core_chain_valid = all([
    len(
        cell126_case_matches
    )
    ==
    1,

    len(
        cell126_task_matches
    )
    ==
    1,

    len(
        cell126_target_matches
    )
    ==
    1,

    cell126_case_task_link,
    cell126_target_case_link,
    cell126_target_task_link,
    cell126_source_type_match,
    cell126_stratum_match,
    cell126_target_parts_present,
    cell126_target_token_match,
    cell126_target_selection_basis_match,
    not cell126_candidate_is_active_pilot,
])


if (
    cell126_core_chain_valid
    and
    cell126_groundability_evidence_found
):

    cell126_chain_status = (
        "EXCLUSION_WINNER_CASE_TASK_TARGET_CHAIN_VALID_"
        "GROUNDABILITY_EVIDENCE_RESOLVED"
    )

    cell126_next_action = (
        "INSPECT_GROUNDABILITY_DECISION_AND_CANDIDATE_TARGET_TEXT_"
        "BEFORE_REPLACEMENT_STAGING"
    )


elif (
    cell126_core_chain_valid
    and
    not cell126_groundability_evidence_found
):

    cell126_chain_status = (
        "EXCLUSION_WINNER_CORE_CHAIN_VALID_"
        "GROUNDABILITY_RECORD_NOT_YET_RESOLVED"
    )

    cell126_next_action = (
        "INSPECT_HISTORICAL_GROUNDABILITY_CONSTRUCTION_"
        "FOR_THIS_CANDIDATE"
    )


else:

    cell126_chain_status = (
        "EXCLUSION_WINNER_CASE_TASK_TARGET_CHAIN_INVALID"
    )

    cell126_next_action = (
        "DO_NOT_STAGE_REPLACEMENT"
    )


# ============================================================
# 14) Result
# ============================================================

cell126_exclusion_winner_chain_inspection = {
    "inspection_version":
        "exclusion_winner_case_task_target_groundability_v1",

    "golden_case_id":
        cell126_candidate_case_id,

    "task_draft_id":
        cell126_candidate_task_id,

    "target_evidence_id":
        cell126_candidate_target_id,

    "case_match_count":
        len(
            cell126_case_matches
        ),

    "task_match_count":
        len(
            cell126_task_matches
        ),

    "target_match_count":
        len(
            cell126_target_matches
        ),

    "case_task_link":
        cell126_case_task_link,

    "target_case_link":
        cell126_target_case_link,

    "target_task_link":
        cell126_target_task_link,

    "source_type_match":
        cell126_source_type_match,

    "stratum_match":
        cell126_stratum_match,

    "target_parts_present":
        cell126_target_parts_present,

    "target_token_match":
        cell126_target_token_match,

    "target_selection_basis_match":
        cell126_target_selection_basis_match,

    "candidate_is_active_pilot":
        cell126_candidate_is_active_pilot,

    "groundability_runtime_object_count":
        len(
            cell126_groundability_runtime_objects
        ),

    "groundability_match_count":
        cell126_groundability_match_count,

    "groundability_evidence_found":
        cell126_groundability_evidence_found,

    "core_chain_valid":
        cell126_core_chain_valid,

    "replacement_staged":
        False,

    "replacement_activated":
        False,

    "queue_advanced":
        False,

    "status":
        cell126_chain_status,

    "next_action":
        cell126_next_action,
}


cell126_exclusion_winner_chain_sha256 = (
    cell126_sha(
        cell126_exclusion_winner_chain_inspection
    )
)


# ============================================================
# 15) Freeze protected state AFTER inspection
# ============================================================

cell126_after = {
    "pilots":
        cell126_sha(
            question_drafting_pilot_packets
        ),

    "pilot_index":
        cell126_sha(
            question_drafting_pilot_by_id
        ),

    "cases":
        cell126_sha(
            golden_case_records
        ),

    "tasks":
        cell126_sha(
            task_draft_plan_rows
        ),

    "targets":
        cell126_sha(
            target_evidence_records
        ),

    "queue":
        cell126_sha(
            cell95_remaining_pilot_queue
        ),

    "question":
        cell126_sha(
            cell97_question_draft_record
        ),
}


# ============================================================
# 16) Gates
# ============================================================

cell126_gates = {
    "cell125_is_pass":
        globals().get(
            "cell_125_pass"
        )
        is True,

    "required_objects_exist":
        len(
            cell126_missing_objects
        )
        ==
        0,

    "candidate_resolved":
        isinstance(
            cell126_candidate,
            dict,
        ),

    "exact_case_resolved":
        len(
            cell126_case_matches
        )
        ==
        1,

    "exact_task_resolved":
        len(
            cell126_task_matches
        )
        ==
        1,

    "exact_target_resolved":
        len(
            cell126_target_matches
        )
        ==
        1,

    "case_task_link_valid":
        cell126_case_task_link,

    "target_case_link_valid":
        cell126_target_case_link,

    "target_task_link_valid":
        cell126_target_task_link,

    "source_type_matches_current_stratum":
        cell126_source_type_match,

    "stratum_matches_current_stratum":
        cell126_stratum_match,

    "target_parts_present":
        cell126_target_parts_present,

    "target_token_count_matches_dry_run":
        cell126_target_token_match,

    "target_selection_basis_matches_dry_run":
        cell126_target_selection_basis_match,

    "candidate_not_already_active_pilot":
        not cell126_candidate_is_active_pilot,

    "groundability_runtime_inspected":
        isinstance(
            cell126_groundability_runtime_objects,
            list,
        ),

    "groundability_resolution_completed":
        isinstance(
            cell126_groundability_evidence_found,
            bool,
        ),

    "core_chain_valid":
        cell126_core_chain_valid,

    "replacement_not_staged":
        cell126_exclusion_winner_chain_inspection[
            "replacement_staged"
        ]
        is False,

    "replacement_not_activated":
        cell126_exclusion_winner_chain_inspection[
            "replacement_activated"
        ]
        is False,

    "queue_not_advanced":
        cell126_exclusion_winner_chain_inspection[
            "queue_advanced"
        ]
        is False,

    "protected_state_byte_stable":
        cell126_before
        ==
        cell126_after,

    "no_artifact_write":
        True,
}


cell_126_pass = all(
    cell126_gates.values()
)


cell126_status = (
    "EXCLUSION_WINNER_CASE_TASK_TARGET_GROUNDABILITY_INSPECTED"

    if cell_126_pass

    else
    "EXCLUSION_WINNER_CASE_TASK_TARGET_GROUNDABILITY_INSPECTION_FAILED"
)


# ============================================================
# 17) Diagnostics
# ============================================================

print(
    "=== EXCLUSION WINNER — FULL CHAIN INSPECTION ==="
)


print(
    "\n=== CANDIDATE IDENTITY ==="
)

print(
    cell126_candidate
)


print(
    "\n=== RECORD CARDINALITY ==="
)

print(
    "Case records:",
    len(
        cell126_case_matches
    )
)

print(
    "Task records:",
    len(
        cell126_task_matches
    )
)

print(
    "Target records:",
    len(
        cell126_target_matches
    )
)


print(
    "\n=== CHAIN LINKS ==="
)

print(
    "Case -> Task:",
    cell126_case_task_link
)

print(
    "Target -> Case:",
    cell126_target_case_link
)

print(
    "Target -> Task:",
    cell126_target_task_link
)

print(
    "Source type match:",
    cell126_source_type_match
)

print(
    "Stratum match:",
    cell126_stratum_match
)

print(
    "Target parts present:",
    cell126_target_parts_present
)

print(
    "Target token match:",
    cell126_target_token_match
)

print(
    "Selection basis match:",
    cell126_target_selection_basis_match
)


print(
    "\n=== CASE STATUS FIELDS ==="
)

print(
    cell126_case_status_view
)


print(
    "\n=== TASK STATUS FIELDS ==="
)

print(
    cell126_task_status_view
)


print(
    "\n=== TARGET STATUS FIELDS ==="
)

print(
    cell126_target_status_view
)


print(
    "\n=== GROUNDABILITY RUNTIME OBJECTS ==="
)

print(
    "Count:",
    len(
        cell126_groundability_runtime_objects
    )
)


for item in (
    cell126_groundability_runtime_objects
):

    print(
        "-",
        item
    )


print(
    "\n=== CANDIDATE GROUNDABILITY MATCHES ==="
)

print(
    "Count:",
    cell126_groundability_match_count
)


for match in (
    cell126_groundability_matches
):

    print(
        "\nRuntime object:",
        match[
            "runtime_object"
        ]
    )

    print(
        "Status fields:",
        match[
            "status_fields"
        ]
    )

    print(
        "Record keys:",
        sorted(
            match[
                "record"
            ].keys()
        )
    )


print(
    "\n=== GROUNDABILITY STATUS VALUES ==="
)

for item in (
    cell126_groundability_status_values
):

    print(
        "-",
        item
    )


print(
    "\n=== INTERPRETATION ==="
)

print(
    "Candidate already active Pilot:",
    cell126_candidate_is_active_pilot
)

print(
    "Core chain valid:",
    cell126_core_chain_valid
)

print(
    "Groundability evidence found:",
    cell126_groundability_evidence_found
)

print(
    "Status:",
    cell126_chain_status
)

print(
    "Next action:",
    cell126_next_action
)

print(
    "Inspection SHA256:",
    cell126_exclusion_winner_chain_sha256
)


print(
    "\n=== FAILURES ==="
)

print(
    "Prerequisite failures:",
    cell126_prerequisite_failures
)


print(
    "\nSTATUS:",
    cell126_status
)


print(
    "\n=== CODE CELL 126 GATES ==="
)


for name, passed in (
    cell126_gates.items()
):

    print(
        name,
        ":",
        (
            "PASS"
            if passed
            else "FAIL"
        ),
    )


print(
    "\nFAILED GATES:",
    [
        name

        for name, passed
        in cell126_gates.items()

        if not passed
    ],
)


print(
    "\nCODE CELL 126 GATE:",
    (
        "PASS"
        if cell_126_pass
        else "FAIL"
    ),
)


assert cell_126_pass, (
    "Exclusion-winner Case/Task/Target/Groundability inspection failed. "
    "Do not stage or activate a replacement. "
    "Repair this SAME Code Cell 126."
)
```

```python
# Code Cell 127

# ============================================================
# Inspect the exclusion winner's actual Evidence text,
# provenance, attribution, and Groundability decision.
#
# Read-only only.
# No replacement staging.
# No question generation.
# No activation.
# No queue advancement.
# No artifact write.
# ============================================================


# ============================================================
# 1) Prerequisites
# ============================================================

cell127_prerequisite_failures = []


if globals().get(
    "cell_126_pass"
) is not True:

    cell127_prerequisite_failures.append(
        "CELL126_NOT_PASS"
    )


if (
    globals().get(
        "cell126_status"
    )
    !=
    "EXCLUSION_WINNER_CASE_TASK_TARGET_GROUNDABILITY_INSPECTED"
):

    cell127_prerequisite_failures.append({
        "reason":
            "CELL126_STATUS_UNEXPECTED",

        "actual":
            globals().get(
                "cell126_status"
            ),
    })


CELL127_REQUIRED_OBJECTS = [
    "cell126_candidate",
    "cell126_target_record",
    "case_groundability_precheck",
    "corpus_target_index",
    "question_drafting_pilot_packets",
    "question_drafting_pilot_by_id",
    "cell95_remaining_pilot_queue",
    "cell97_question_draft_record",
    "target_evidence_records",
]


cell127_missing_objects = [
    name

    for name
    in CELL127_REQUIRED_OBJECTS

    if name not in globals()
]


if cell127_missing_objects:

    cell127_prerequisite_failures.append({
        "reason":
            "REQUIRED_RUNTIME_OBJECTS_MISSING",

        "objects":
            cell127_missing_objects,
    })


# ============================================================
# 2) Helper
# ============================================================

def cell127_sha(value):

    return hashlib.sha256(
        stable_json_key(
            value
        ).encode(
            "utf-8"
        )
    ).hexdigest()


# ============================================================
# 3) Freeze protected state BEFORE inspection
# ============================================================

cell127_before = {
    "pilots":
        cell127_sha(
            question_drafting_pilot_packets
        ),

    "pilot_index":
        cell127_sha(
            question_drafting_pilot_by_id
        ),

    "targets":
        cell127_sha(
            target_evidence_records
        ),

    "queue":
        cell127_sha(
            cell95_remaining_pilot_queue
        ),

    "question":
        cell127_sha(
            cell97_question_draft_record
        ),
}


# ============================================================
# 4) Resolve Candidate identity
# ============================================================

cell127_case_id = (
    cell126_candidate.get(
        "golden_case_id"
    )
)


cell127_task_id = (
    cell126_candidate.get(
        "task_draft_id"
    )
)


cell127_target_id = (
    cell126_candidate.get(
        "target_evidence_id"
    )
)


cell127_target_parts = (
    cell126_target_record.get(
        "target_parts"
    )

    if isinstance(
        cell126_target_record,
        dict,
    )

    else None
)


# ============================================================
# 5) Resolve canonical Groundability row
# ============================================================

cell127_groundability_matches = [
    record

    for record
    in case_groundability_precheck

    if (
        isinstance(
            record,
            dict,
        )

        and

        record.get(
            "golden_case_id"
        )
        ==
        cell127_case_id
    )
]


cell127_groundability_record = (
    cell127_groundability_matches[
        0
    ]

    if len(
        cell127_groundability_matches
    )
    ==
    1

    else None
)


cell127_groundability_status = (
    cell127_groundability_record.get(
        "groundability_precheck_status"
    )

    if isinstance(
        cell127_groundability_record,
        dict,
    )

    else None
)


cell127_evidence_text_recoverable = (
    cell127_groundability_record.get(
        "evidence_text_recoverable"
    )

    if isinstance(
        cell127_groundability_record,
        dict,
    )

    else None
)


cell127_external_knowledge_required = (
    cell127_groundability_record.get(
        "external_knowledge_required"
    )

    if isinstance(
        cell127_groundability_record,
        dict,
    )

    else None
)


cell127_construction_eligibility = (
    cell127_groundability_record.get(
        "construction_eligibility"
    )

    if isinstance(
        cell127_groundability_record,
        dict,
    )

    else None
)


cell127_unknown_speaker_inference = (
    cell127_groundability_record.get(
        "unknown_speaker_inference"
    )

    if isinstance(
        cell127_groundability_record,
        dict,
    )

    else None
)


# ============================================================
# 6) Recover exact Target text from Accepted Corpus
# ============================================================

cell127_recovery_failures = []
cell127_recovered_parts = []


if isinstance(
    cell127_target_parts,
    list,
):

    for index, part in enumerate(
        cell127_target_parts
    ):

        if not isinstance(
            part,
            dict,
        ):

            cell127_recovery_failures.append({
                "part_index":
                    index,

                "reason":
                    "TARGET_PART_NOT_DICT",
            })

            continue


        document_id = (
            part.get(
                "document_id"
            )
        )


        char_start = (
            part.get(
                "char_start"
            )
        )


        char_end = (
            part.get(
                "char_end"
            )
        )


        stored_text_sha256 = (
            part.get(
                "target_text_sha256"
            )
        )


        corpus_record = (
            corpus_target_index.get(
                document_id
            )

            if isinstance(
                corpus_target_index,
                dict,
            )

            else None
        )


        if not isinstance(
            corpus_record,
            dict,
        ):

            cell127_recovery_failures.append({
                "part_index":
                    index,

                "reason":
                    "CORPUS_DOCUMENT_NOT_RESOLVED",

                "document_id":
                    document_id,
            })

            continue


        corpus_text = (
            corpus_record.get(
                "text"
            )
        )


        if not (
            isinstance(
                corpus_text,
                str,
            )

            and

            isinstance(
                char_start,
                int,
            )

            and

            isinstance(
                char_end,
                int,
            )

            and

            0
            <=
            char_start
            <=
            char_end
            <=
            len(
                corpus_text
            )
        ):

            cell127_recovery_failures.append({
                "part_index":
                    index,

                "reason":
                    "INVALID_TEXT_SPAN",
            })

            continue


        recovered_text = (
            corpus_text[
                char_start:
                char_end
            ]
        )


        recovered_sha256 = hashlib.sha256(
            recovered_text.encode(
                "utf-8"
            )
        ).hexdigest()


        locator = (
            part.get(
                "locator"
            )
        )


        attribution = (
            part.get(
                "attribution"
            )
        )


        locator_present = (
            isinstance(
                locator,
                dict,
            )

            and

            len(
                locator
            )
            >
            0
        )


        attribution_present = (
            isinstance(
                attribution,
                dict,
            )

            and

            len(
                attribution
            )
            >
            0
        )


        attribution_policy_validated = (
            attribution_present

            and

            attribution.get(
                "attribution_status"
            )
            ==
            "policy_validated"
        )


        hash_match = (
            recovered_sha256
            ==
            stored_text_sha256
        )


        cell127_recovered_parts.append({
            "part_index":
                index,

            "document_id":
                document_id,

            "role":
                part.get(
                    "role"
                ),

            "char_start":
                char_start,

            "char_end":
                char_end,

            "token_count":
                part.get(
                    "token_count"
                ),

            "stored_text_sha256":
                stored_text_sha256,

            "recovered_text_sha256":
                recovered_sha256,

            "hash_match":
                hash_match,

            "locator_present":
                locator_present,

            "locator":
                locator,

            "attribution_present":
                attribution_present,

            "attribution_policy_validated":
                attribution_policy_validated,

            "attribution":
                attribution,

            "recovered_text":
                recovered_text,
        })


# ============================================================
# 7) Evidence integrity decision
# ============================================================

cell127_part_count_matches = (
    isinstance(
        cell127_target_parts,
        list,
    )

    and

    len(
        cell127_target_parts
    )
    ==
    len(
        cell127_recovered_parts
    )

    and

    len(
        cell127_recovered_parts
    )
    >
    0
)


cell127_all_hashes_match = (
    cell127_part_count_matches

    and

    all(
        part[
            "hash_match"
        ]

        for part
        in cell127_recovered_parts
    )
)


cell127_all_locators_present = (
    cell127_part_count_matches

    and

    all(
        part[
            "locator_present"
        ]

        for part
        in cell127_recovered_parts
    )
)


cell127_all_attribution_validated = (
    cell127_part_count_matches

    and

    all(
        part[
            "attribution_policy_validated"
        ]

        for part
        in cell127_recovered_parts
    )
)


# ============================================================
# 8) Groundability decision
# ============================================================

cell127_groundability_pass = (
    cell127_groundability_status
    ==
    "EVIDENCE_RECOVERABILITY_PASS"
)


cell127_external_knowledge_forbidden_satisfied = (
    cell127_external_knowledge_required
    is False
)


cell127_recoverability_flag_pass = (
    cell127_evidence_text_recoverable
    is True
)


# ============================================================
# 9) Cross-check duplicate Groundability evidence
#    previously discovered in Cell 126
# ============================================================

cell127_duplicate_groundability_statuses = []


for match in globals().get(
    "cell126_groundability_matches",
    [],
):

    if not isinstance(
        match,
        dict,
    ):
        continue


    record = (
        match.get(
            "record"
        )
    )


    if not isinstance(
        record,
        dict,
    ):
        continue


    cell127_duplicate_groundability_statuses.append({
        "runtime_object":
            match.get(
                "runtime_object"
            ),

        "groundability_precheck_status":
            record.get(
                "groundability_precheck_status"
            ),

        "evidence_text_recoverable":
            record.get(
                "evidence_text_recoverable"
            ),

        "external_knowledge_required":
            record.get(
                "external_knowledge_required"
            ),

        "construction_eligibility":
            record.get(
                "construction_eligibility"
            ),
    })


cell127_groundability_duplicates_consistent = (
    len(
        cell127_duplicate_groundability_statuses
    )
    >
    0

    and

    len({
        stable_json_key(
            {
                "groundability_precheck_status":
                    item[
                        "groundability_precheck_status"
                    ],

                "evidence_text_recoverable":
                    item[
                        "evidence_text_recoverable"
                    ],

                "external_knowledge_required":
                    item[
                        "external_knowledge_required"
                    ],

                "construction_eligibility":
                    item[
                        "construction_eligibility"
                    ],
            }
        )

        for item
        in cell127_duplicate_groundability_statuses
    })
    ==
    1
)


# ============================================================
# 10) Conservative pre-staging readiness
# ============================================================

cell127_pre_staging_evidence_ready = all([
    len(
        cell127_groundability_matches
    )
    ==
    1,

    cell127_groundability_pass,

    cell127_recoverability_flag_pass,

    cell127_external_knowledge_forbidden_satisfied,

    cell127_all_hashes_match,

    cell127_all_locators_present,

    cell127_all_attribution_validated,

    cell127_groundability_duplicates_consistent,

    len(
        cell127_recovery_failures
    )
    ==
    0,
])


if cell127_pre_staging_evidence_ready:

    cell127_readiness_status = (
        "EXCLUSION_WINNER_EVIDENCE_AND_GROUNDABILITY_"
        "VALIDATED_FOR_PRE_STAGING_REVIEW"
    )

    cell127_next_action = (
        "SIMULATE_REPLACEMENT_PILOT_PACKET_WITHOUT_ACTIVATION_"
        "AND_VALIDATE_SCHEMA_LINEAGE"
    )


else:

    cell127_readiness_status = (
        "EXCLUSION_WINNER_NOT_READY_FOR_REPLACEMENT_STAGING"
    )

    cell127_next_action = (
        "DO_NOT_STAGE_REPLACEMENT"
    )


# ============================================================
# 11) Result record
# ============================================================

cell127_candidate_evidence_review = {
    "inspection_version":
        "exclusion_winner_evidence_groundability_review_v1",

    "golden_case_id":
        cell127_case_id,

    "task_draft_id":
        cell127_task_id,

    "target_evidence_id":
        cell127_target_id,

    "groundability_match_count":
        len(
            cell127_groundability_matches
        ),

    "groundability_precheck_status":
        cell127_groundability_status,

    "evidence_text_recoverable":
        cell127_evidence_text_recoverable,

    "external_knowledge_required":
        cell127_external_knowledge_required,

    "construction_eligibility":
        cell127_construction_eligibility,

    "unknown_speaker_inference":
        cell127_unknown_speaker_inference,

    "target_part_count":
        (
            len(
                cell127_target_parts
            )

            if isinstance(
                cell127_target_parts,
                list,
            )

            else None
        ),

    "all_hashes_match":
        cell127_all_hashes_match,

    "all_locators_present":
        cell127_all_locators_present,

    "all_attribution_validated":
        cell127_all_attribution_validated,

    "groundability_duplicates_consistent":
        cell127_groundability_duplicates_consistent,

    "pre_staging_evidence_ready":
        cell127_pre_staging_evidence_ready,

    "replacement_staged":
        False,

    "replacement_activated":
        False,

    "question_generated":
        False,

    "queue_advanced":
        False,

    "status":
        cell127_readiness_status,

    "next_action":
        cell127_next_action,
}


cell127_candidate_evidence_review_sha256 = (
    cell127_sha(
        cell127_candidate_evidence_review
    )
)


# ============================================================
# 12) Freeze protected state AFTER inspection
# ============================================================

cell127_after = {
    "pilots":
        cell127_sha(
            question_drafting_pilot_packets
        ),

    "pilot_index":
        cell127_sha(
            question_drafting_pilot_by_id
        ),

    "targets":
        cell127_sha(
            target_evidence_records
        ),

    "queue":
        cell127_sha(
            cell95_remaining_pilot_queue
        ),

    "question":
        cell127_sha(
            cell97_question_draft_record
        ),
}


# ============================================================
# 13) Gates
# ============================================================

cell127_gates = {
    "cell126_is_pass":
        globals().get(
            "cell_126_pass"
        )
        is True,

    "required_objects_exist":
        len(
            cell127_missing_objects
        )
        ==
        0,

    "canonical_groundability_record_resolved":
        len(
            cell127_groundability_matches
        )
        ==
        1,

    "groundability_precheck_pass":
        cell127_groundability_pass,

    "evidence_text_recoverable":
        cell127_recoverability_flag_pass,

    "external_knowledge_not_required":
        cell127_external_knowledge_forbidden_satisfied,

    "target_parts_recovered":
        cell127_part_count_matches,

    "all_target_text_hashes_match":
        cell127_all_hashes_match,

    "all_target_locators_present":
        cell127_all_locators_present,

    "all_target_attribution_policy_validated":
        cell127_all_attribution_validated,

    "groundability_runtime_duplicates_consistent":
        cell127_groundability_duplicates_consistent,

    "recovery_has_no_failures":
        len(
            cell127_recovery_failures
        )
        ==
        0,

    "pre_staging_evidence_ready":
        cell127_pre_staging_evidence_ready,

    "replacement_not_staged":
        cell127_candidate_evidence_review[
            "replacement_staged"
        ]
        is False,

    "replacement_not_activated":
        cell127_candidate_evidence_review[
            "replacement_activated"
        ]
        is False,

    "question_not_generated":
        cell127_candidate_evidence_review[
            "question_generated"
        ]
        is False,

    "queue_not_advanced":
        cell127_candidate_evidence_review[
            "queue_advanced"
        ]
        is False,

    "protected_state_byte_stable":
        cell127_before
        ==
        cell127_after,

    "no_artifact_write":
        True,
}


cell_127_pass = all(
    cell127_gates.values()
)


cell127_status = (
    "EXCLUSION_WINNER_EVIDENCE_AND_GROUNDABILITY_INSPECTED"

    if cell_127_pass

    else
    "EXCLUSION_WINNER_EVIDENCE_AND_GROUNDABILITY_INSPECTION_FAILED"
)


# ============================================================
# 14) Diagnostics
# ============================================================

print(
    "=== EXCLUSION WINNER — EVIDENCE + GROUNDABILITY REVIEW ==="
)


print(
    "\n=== IDENTITY ==="
)

print(
    "Golden Case:",
    cell127_case_id
)

print(
    "Task Draft:",
    cell127_task_id
)

print(
    "Target Evidence:",
    cell127_target_id
)


print(
    "\n=== GROUNDABILITY DECISION ==="
)

print(
    "Precheck status:",
    cell127_groundability_status
)

print(
    "Evidence text recoverable:",
    cell127_evidence_text_recoverable
)

print(
    "External knowledge required:",
    cell127_external_knowledge_required
)

print(
    "Construction eligibility:",
    cell127_construction_eligibility
)

print(
    "Unknown-speaker inference:",
    cell127_unknown_speaker_inference
)


print(
    "\n=== RECOVERED TARGET EVIDENCE ==="
)


for part in (
    cell127_recovered_parts
):

    print(
        "\nPart:",
        part[
            "part_index"
        ]
    )

    print(
        "Document:",
        part[
            "document_id"
        ]
    )

    print(
        "Role:",
        part[
            "role"
        ]
    )

    print(
        "Span:",
        (
            part[
                "char_start"
            ],
            part[
                "char_end"
            ],
        )
    )

    print(
        "Token count:",
        part[
            "token_count"
        ]
    )

    print(
        "Hash match:",
        part[
            "hash_match"
        ]
    )

    print(
        "Locator present:",
        part[
            "locator_present"
        ]
    )

    print(
        "Attribution validated:",
        part[
            "attribution_policy_validated"
        ]
    )

    print(
        "Locator:",
        part[
            "locator"
        ]
    )

    print(
        "Attribution:",
        part[
            "attribution"
        ]
    )

    print(
        "\nTEXT:"
    )

    print(
        part[
            "recovered_text"
        ]
    )


print(
    "\n=== GROUNDABILITY CONSISTENCY ==="
)

print(
    "Duplicate runtime records:",
    len(
        cell127_duplicate_groundability_statuses
    )
)

print(
    "Consistent:",
    cell127_groundability_duplicates_consistent
)


for item in (
    cell127_duplicate_groundability_statuses
):

    print(
        "-",
        item
    )


print(
    "\n=== INTERPRETATION ==="
)

print(
    "All hashes match:",
    cell127_all_hashes_match
)

print(
    "All locators present:",
    cell127_all_locators_present
)

print(
    "All attribution validated:",
    cell127_all_attribution_validated
)

print(
    "Pre-staging Evidence ready:",
    cell127_pre_staging_evidence_ready
)

print(
    "Status:",
    cell127_readiness_status
)

print(
    "Next action:",
    cell127_next_action
)

print(
    "Inspection SHA256:",
    cell127_candidate_evidence_review_sha256
)


print(
    "\n=== FAILURES ==="
)

print(
    "Prerequisite failures:",
    cell127_prerequisite_failures
)

print(
    "Recovery failures:",
    cell127_recovery_failures
)


print(
    "\nSTATUS:",
    cell127_status
)


print(
    "\n=== CODE CELL 127 GATES ==="
)


for name, passed in (
    cell127_gates.items()
):

    print(
        name,
        ":",
        (
            "PASS"
            if passed
            else "FAIL"
        ),
    )


print(
    "\nFAILED GATES:",
    [
        name

        for name, passed
        in cell127_gates.items()

        if not passed
    ],
)


print(
    "\nCODE CELL 127 GATE:",
    (
        "PASS"
        if cell_127_pass
        else "FAIL"
    ),
)


assert cell_127_pass, (
    "Exclusion-winner Evidence/Groundability inspection failed. "
    "Do not stage or activate a replacement. "
    "Repair this SAME Code Cell 127."
)
```

```python
# Code Cell 128

# ============================================================
# Resolve the EXACT historical Pilot-packet schema and
# pilot_case_id construction expression before simulation.
#
# Read-only only.
# No replacement staging.
# No activation.
# No queue advancement.
# No artifact write.
# ============================================================

import ast
import textwrap


# ============================================================
# 1) Prerequisites
# ============================================================

cell128_prerequisite_failures = []


if globals().get(
    "cell_127_pass"
) is not True:

    cell128_prerequisite_failures.append(
        "CELL127_NOT_PASS"
    )


if (
    globals().get(
        "cell127_status"
    )
    !=
    "EXCLUSION_WINNER_EVIDENCE_AND_GROUNDABILITY_INSPECTED"
):

    cell128_prerequisite_failures.append({
        "reason":
            "CELL127_STATUS_UNEXPECTED",

        "actual":
            globals().get(
                "cell127_status"
            ),
    })


CELL128_REQUIRED_OBJECTS = [
    "cell125_history29_source",
    "cell125_exclusion_candidate_summary",
    "question_drafting_pilot_by_id",
    "question_drafting_pilot_packets",
    "cell97_question_draft_record",
    "cell95_remaining_pilot_queue",
]


cell128_missing_objects = [
    name

    for name
    in CELL128_REQUIRED_OBJECTS

    if name not in globals()
]


if cell128_missing_objects:

    cell128_prerequisite_failures.append({
        "reason":
            "REQUIRED_RUNTIME_OBJECTS_MISSING",

        "objects":
            cell128_missing_objects,
    })


# ============================================================
# 2) Helper
# ============================================================

def cell128_sha(value):

    return hashlib.sha256(
        stable_json_key(
            value
        ).encode(
            "utf-8"
        )
    ).hexdigest()


# ============================================================
# 3) Freeze protected state
# ============================================================

cell128_before = {
    "pilots":
        cell128_sha(
            question_drafting_pilot_packets
        ),

    "pilot_index":
        cell128_sha(
            question_drafting_pilot_by_id
        ),

    "queue":
        cell128_sha(
            cell95_remaining_pilot_queue
        ),

    "question":
        cell128_sha(
            cell97_question_draft_record
        ),
}


# ============================================================
# 4) Resolve current Pilot packet
# ============================================================

cell128_current_pilot_id = (
    cell97_question_draft_record.get(
        "pilot_case_id"
    )
)


cell128_current_packet = (
    question_drafting_pilot_by_id.get(
        cell128_current_pilot_id
    )
)


cell128_current_packet_keys = (
    sorted(
        cell128_current_packet.keys()
    )

    if isinstance(
        cell128_current_packet,
        dict,
    )

    else []
)


# ============================================================
# 5) Parse exact Cell 29 source
# ============================================================

cell128_ast_failures = []


try:

    cell128_source = textwrap.dedent(
        cell125_history29_source
    )


    cell128_tree = ast.parse(
        cell128_source
    )


except Exception as exc:

    cell128_source = None
    cell128_tree = None

    cell128_ast_failures.append({
        "reason":
            "CELL29_AST_PARSE_FAILED",

        "error":
            repr(
                exc
            ),
    })


# ============================================================
# 6) Find Pilot-packet Dict candidates
# ============================================================

CELL128_CORE_PACKET_KEYS = {
    "pilot_case_id",
    "golden_case_id",
    "task_draft_id",
    "target_evidence_id",
    "source_type",
    "stratum",
}


cell128_packet_dict_candidates = []


if cell128_tree is not None:

    for node in ast.walk(
        cell128_tree
    ):

        if not isinstance(
            node,
            ast.Dict,
        ):

            continue


        literal_keys = []


        for key_node in node.keys:

            if (
                isinstance(
                    key_node,
                    ast.Constant,
                )

                and

                isinstance(
                    key_node.value,
                    str,
                )
            ):

                literal_keys.append(
                    key_node.value
                )


        key_set = set(
            literal_keys
        )


        if not CELL128_CORE_PACKET_KEYS.issubset(
            key_set
        ):

            continue


        cell128_packet_dict_candidates.append({
            "node":
                node,

            "keys":
                sorted(
                    key_set
                ),

            "exact_current_schema":
                sorted(
                    key_set
                )
                ==
                cell128_current_packet_keys,
        })


# ============================================================
# 7) Resolve exact-schema packet Dict
# ============================================================

cell128_exact_schema_candidates = [
    item

    for item
    in cell128_packet_dict_candidates

    if item[
        "exact_current_schema"
    ]
]


cell128_exact_packet_resolved = (
    len(
        cell128_exact_schema_candidates
    )
    ==
    1
)


cell128_packet_node = (
    cell128_exact_schema_candidates[
        0
    ][
        "node"
    ]

    if cell128_exact_packet_resolved

    else None
)


cell128_historical_packet_keys = (
    cell128_exact_schema_candidates[
        0
    ][
        "keys"
    ]

    if cell128_exact_packet_resolved

    else []
)


# ============================================================
# 8) Extract exact packet source
# ============================================================

cell128_packet_source = (
    ast.get_source_segment(
        cell128_source,
        cell128_packet_node,
    )

    if (
        isinstance(
            cell128_source,
            str,
        )

        and

        cell128_packet_node
        is not None
    )

    else None
)


# ============================================================
# 9) Resolve pilot_case_id expression
# ============================================================

cell128_pilot_id_expression_node = None
cell128_pilot_id_expression_source = None


if (
    cell128_packet_node
    is not None
):

    for key_node, value_node in zip(
        cell128_packet_node.keys,
        cell128_packet_node.values,
    ):

        if (
            isinstance(
                key_node,
                ast.Constant,
            )

            and

            key_node.value
            ==
            "pilot_case_id"
        ):

            cell128_pilot_id_expression_node = (
                value_node
            )

            break


if (
    cell128_pilot_id_expression_node
    is not None
):

    cell128_pilot_id_expression_source = (
        ast.get_source_segment(
            cell128_source,
            cell128_pilot_id_expression_node,
        )
    )


# ============================================================
# 10) Inspect names used by Pilot-ID expression
# ============================================================

cell128_pilot_id_loaded_names = sorted({
    node.id

    for node
    in ast.walk(
        cell128_pilot_id_expression_node
    )

    if (
        isinstance(
            node,
            ast.Name,
        )

        and

        isinstance(
            node.ctx,
            ast.Load,
        )
    )
}) if (
    cell128_pilot_id_expression_node
    is not None
) else []


cell128_pilot_id_runtime_callables = []


for name in (
    cell128_pilot_id_loaded_names
):

    if (
        name in globals()

        and

        callable(
            globals()[
                name
            ]
        )
    ):

        value = globals()[
            name
        ]


        try:

            signature = str(
                inspect.signature(
                    value
                )
            )


        except Exception:

            signature = None


        cell128_pilot_id_runtime_callables.append({
            "name":
                name,

            "signature":
                signature,
        })


# ============================================================
# 11) Extract packet-field expressions
# ============================================================

cell128_packet_field_expressions = {}


if (
    cell128_packet_node
    is not None
):

    for key_node, value_node in zip(
        cell128_packet_node.keys,
        cell128_packet_node.values,
    ):

        if not (
            isinstance(
                key_node,
                ast.Constant,
            )

            and

            isinstance(
                key_node.value,
                str,
            )
        ):

            continue


        cell128_packet_field_expressions[
            key_node.value
        ] = ast.get_source_segment(
            cell128_source,
            value_node,
        )


# ============================================================
# 12) Required replacement identity fields
# ============================================================

CELL128_REPLACEMENT_IDENTITY_FIELDS = [
    "pilot_case_id",
    "golden_case_id",
    "task_draft_id",
    "target_evidence_id",
    "source_type",
    "stratum",
]


cell128_identity_expressions_present = all(
    field
    in
    cell128_packet_field_expressions

    for field
    in CELL128_REPLACEMENT_IDENTITY_FIELDS
)


# ============================================================
# 13) Candidate identity envelope
#
# This is NOT a Pilot packet.
# It is only the exact candidate identity that the next
# simulation must inject into the discovered constructor.
# ============================================================

cell128_candidate_identity_envelope = {
    "golden_case_id":
        cell125_exclusion_candidate_summary.get(
            "golden_case_id"
        ),

    "task_draft_id":
        cell125_exclusion_candidate_summary.get(
            "task_draft_id"
        ),

    "target_evidence_id":
        cell125_exclusion_candidate_summary.get(
            "target_evidence_id"
        ),

    "source_type":
        globals().get(
            "cell125_current_source_type"
        ),

    "stratum":
        globals().get(
            "cell125_current_stratum"
        ),
}


# ============================================================
# 14) Simulation readiness
# ============================================================

cell128_schema_exact_match = (
    cell128_historical_packet_keys
    ==
    cell128_current_packet_keys
)


cell128_pilot_id_expression_resolved = (
    isinstance(
        cell128_pilot_id_expression_source,
        str,
    )

    and

    len(
        cell128_pilot_id_expression_source
    )
    >
    0
)


cell128_simulation_ready = all([
    cell128_exact_packet_resolved,
    cell128_schema_exact_match,
    cell128_pilot_id_expression_resolved,
    cell128_identity_expressions_present,
    isinstance(
        cell128_candidate_identity_envelope[
            "golden_case_id"
        ],
        str,
    ),
    isinstance(
        cell128_candidate_identity_envelope[
            "task_draft_id"
        ],
        str,
    ),
    isinstance(
        cell128_candidate_identity_envelope[
            "target_evidence_id"
        ],
        str,
    ),
])


if cell128_simulation_ready:

    cell128_status_detail = (
        "EXACT_PILOT_PACKET_SCHEMA_AND_ID_CONSTRUCTION_RESOLVED"
    )

    cell128_next_action = (
        "SIMULATE_EXCLUSION_WINNER_PILOT_PACKET_USING_"
        "EXACT_DISCOVERED_CONSTRUCTION_WITHOUT_ACTIVATION"
    )


else:

    cell128_status_detail = (
        "PILOT_PACKET_CONSTRUCTION_NOT_FULLY_RESOLVED"
    )

    cell128_next_action = (
        "DO_NOT_SIMULATE_OR_STAGE_REPLACEMENT"
    )


# ============================================================
# 15) Result
# ============================================================

cell128_pilot_packet_construction_contract = {
    "inspection_version":
        "pilot_packet_construction_contract_v1",

    "current_pilot_id":
        cell128_current_pilot_id,

    "packet_dict_candidate_count":
        len(
            cell128_packet_dict_candidates
        ),

    "exact_schema_candidate_count":
        len(
            cell128_exact_schema_candidates
        ),

    "schema_exact_match":
        cell128_schema_exact_match,

    "current_packet_keys":
        cell128_current_packet_keys,

    "historical_packet_keys":
        cell128_historical_packet_keys,

    "pilot_id_expression":
        cell128_pilot_id_expression_source,

    "pilot_id_loaded_names":
        cell128_pilot_id_loaded_names,

    "pilot_id_runtime_callables":
        cell128_pilot_id_runtime_callables,

    "identity_expressions_present":
        cell128_identity_expressions_present,

    "candidate_identity_envelope":
        cell128_candidate_identity_envelope,

    "simulation_ready":
        cell128_simulation_ready,

    "replacement_packet_created":
        False,

    "replacement_staged":
        False,

    "replacement_activated":
        False,

    "queue_advanced":
        False,

    "status":
        cell128_status_detail,

    "next_action":
        cell128_next_action,
}


cell128_pilot_packet_construction_sha256 = (
    cell128_sha(
        cell128_pilot_packet_construction_contract
    )
)


# ============================================================
# 16) Freeze protected state AFTER inspection
# ============================================================

cell128_after = {
    "pilots":
        cell128_sha(
            question_drafting_pilot_packets
        ),

    "pilot_index":
        cell128_sha(
            question_drafting_pilot_by_id
        ),

    "queue":
        cell128_sha(
            cell95_remaining_pilot_queue
        ),

    "question":
        cell128_sha(
            cell97_question_draft_record
        ),
}


# ============================================================
# 17) Gates
# ============================================================

cell128_gates = {
    "cell127_is_pass":
        globals().get(
            "cell_127_pass"
        )
        is True,

    "required_objects_exist":
        len(
            cell128_missing_objects
        )
        ==
        0,

    "cell29_ast_parsed":
        len(
            cell128_ast_failures
        )
        ==
        0,

    "exact_packet_dict_resolved":
        cell128_exact_packet_resolved,

    "historical_schema_matches_current_schema":
        cell128_schema_exact_match,

    "pilot_id_expression_resolved":
        cell128_pilot_id_expression_resolved,

    "identity_field_expressions_present":
        cell128_identity_expressions_present,

    "candidate_identity_envelope_complete":
        all(
            isinstance(
                cell128_candidate_identity_envelope.get(
                    field
                ),
                str,
            )

            for field
            in [
                "golden_case_id",
                "task_draft_id",
                "target_evidence_id",
                "source_type",
                "stratum",
            ]
        ),

    "simulation_ready":
        cell128_simulation_ready,

    "replacement_packet_not_created":
        cell128_pilot_packet_construction_contract[
            "replacement_packet_created"
        ]
        is False,

    "replacement_not_staged":
        cell128_pilot_packet_construction_contract[
            "replacement_staged"
        ]
        is False,

    "replacement_not_activated":
        cell128_pilot_packet_construction_contract[
            "replacement_activated"
        ]
        is False,

    "queue_not_advanced":
        cell128_pilot_packet_construction_contract[
            "queue_advanced"
        ]
        is False,

    "protected_state_byte_stable":
        cell128_before
        ==
        cell128_after,

    "no_artifact_write":
        True,
}


cell_128_pass = all(
    cell128_gates.values()
)


cell128_status = (
    "PILOT_PACKET_CONSTRUCTION_SCHEMA_AND_ID_INSPECTED"

    if cell_128_pass

    else
    "PILOT_PACKET_CONSTRUCTION_SCHEMA_AND_ID_INSPECTION_FAILED"
)


# ============================================================
# 18) Diagnostics
# ============================================================

print(
    "=== PILOT PACKET CONSTRUCTION CONTRACT ==="
)


print(
    "\n=== SCHEMA ==="
)

print(
    "Packet Dict candidates:",
    len(
        cell128_packet_dict_candidates
    )
)

print(
    "Exact-schema candidates:",
    len(
        cell128_exact_schema_candidates
    )
)

print(
    "Schema exact match:",
    cell128_schema_exact_match
)

print(
    "Current keys:",
    cell128_current_packet_keys
)

print(
    "Historical keys:",
    cell128_historical_packet_keys
)


print(
    "\n=== PILOT ID CONSTRUCTION ==="
)

print(
    "Expression:",
    cell128_pilot_id_expression_source
)

print(
    "Loaded names:",
    cell128_pilot_id_loaded_names
)

print(
    "Runtime callables:",
    cell128_pilot_id_runtime_callables
)


print(
    "\n=== IDENTITY FIELD EXPRESSIONS ==="
)

for field in (
    CELL128_REPLACEMENT_IDENTITY_FIELDS
):

    print(
        field,
        ":",
        cell128_packet_field_expressions.get(
            field
        ),
    )


print(
    "\n=== CANDIDATE IDENTITY ENVELOPE ==="
)

print(
    cell128_candidate_identity_envelope
)


print(
    "\n=== INTERPRETATION ==="
)

print(
    "Simulation ready:",
    cell128_simulation_ready
)

print(
    "Status:",
    cell128_status_detail
)

print(
    "Next action:",
    cell128_next_action
)

print(
    "Inspection SHA256:",
    cell128_pilot_packet_construction_sha256
)


print(
    "\n=== FAILURES ==="
)

print(
    "Prerequisite failures:",
    cell128_prerequisite_failures
)

print(
    "AST failures:",
    cell128_ast_failures
)


print(
    "\nSTATUS:",
    cell128_status
)


print(
    "\n=== CODE CELL 128 GATES ==="
)


for name, passed in (
    cell128_gates.items()
):

    print(
        name,
        ":",
        (
            "PASS"
            if passed
            else "FAIL"
        ),
    )


print(
    "\nFAILED GATES:",
    [
        name

        for name, passed
        in cell128_gates.items()

        if not passed
    ],
)


print(
    "\nCODE CELL 128 GATE:",
    (
        "PASS"
        if cell_128_pass
        else "FAIL"
    ),
)


assert cell_128_pass, (
    "Pilot-packet construction schema/ID inspection failed. "
    "Do not simulate, stage, or activate a replacement. "
    "Repair this SAME Code Cell 128."
)
```

```python
# Code Cell 129

# ============================================================
# SAME-CELL REPAIR
#
# Resolve the exact pilot_case_id assignment and all packet
# dependencies, including:
#   - normal assignments
#   - for-loop target bindings
#   - Python built-ins
#
# Read-only only.
# No replacement packet creation.
# No staging.
# No activation.
# No queue advancement.
# No artifact write.
# ============================================================

import ast
import builtins
import inspect


# ============================================================
# 1) Prerequisites
# ============================================================

cell129_prerequisite_failures = []


if globals().get(
    "cell_128_pass"
) is not True:

    cell129_prerequisite_failures.append(
        "CELL128_NOT_PASS"
    )


if (
    globals().get(
        "cell128_status"
    )
    !=
    "PILOT_PACKET_CONSTRUCTION_SCHEMA_AND_ID_INSPECTED"
):

    cell129_prerequisite_failures.append({
        "reason":
            "CELL128_STATUS_UNEXPECTED",

        "actual":
            globals().get(
                "cell128_status"
            ),
    })


CELL129_REQUIRED_OBJECTS = [
    "cell128_source",
    "cell128_tree",
    "cell128_packet_node",
    "cell128_packet_field_expressions",
    "cell128_candidate_identity_envelope",
    "question_drafting_pilot_packets",
    "question_drafting_pilot_by_id",
    "cell95_remaining_pilot_queue",
    "cell97_question_draft_record",
]


cell129_missing_objects = [
    name

    for name
    in CELL129_REQUIRED_OBJECTS

    if name not in globals()
]


if cell129_missing_objects:

    cell129_prerequisite_failures.append({
        "reason":
            "REQUIRED_RUNTIME_OBJECTS_MISSING",

        "objects":
            cell129_missing_objects,
    })


# ============================================================
# 2) Helpers
# ============================================================

def cell129_sha(value):

    return hashlib.sha256(
        stable_json_key(
            value
        ).encode(
            "utf-8"
        )
    ).hexdigest()


def cell129_loaded_names(node):

    if node is None:
        return []


    return sorted({
        child.id

        for child
        in ast.walk(
            node
        )

        if (
            isinstance(
                child,
                ast.Name,
            )

            and

            isinstance(
                child.ctx,
                ast.Load,
            )
        )
    })


def cell129_bound_names(target):

    names = []


    if isinstance(
        target,
        ast.Name,
    ):

        names.append(
            target.id
        )


    elif isinstance(
        target,
        (ast.Tuple, ast.List),
    ):

        for child in target.elts:

            names.extend(
                cell129_bound_names(
                    child
                )
            )


    return names


# ============================================================
# 3) Freeze protected state BEFORE inspection
# ============================================================

cell129_before = {
    "pilots":
        cell129_sha(
            question_drafting_pilot_packets
        ),

    "pilot_index":
        cell129_sha(
            question_drafting_pilot_by_id
        ),

    "queue":
        cell129_sha(
            cell95_remaining_pilot_queue
        ),

    "question":
        cell129_sha(
            cell97_question_draft_record
        ),
}


# ============================================================
# 4) Packet position
# ============================================================

cell129_packet_lineno = getattr(
    cell128_packet_node,
    "lineno",
    None,
)


# ============================================================
# 5) Collect prior Assign bindings
# ============================================================

cell129_assignment_bindings = {}


if (
    cell128_tree
    is not None

    and

    isinstance(
        cell129_packet_lineno,
        int,
    )
):

    for node in ast.walk(
        cell128_tree
    ):

        if not isinstance(
            node,
            (
                ast.Assign,
                ast.AnnAssign,
            ),
        ):

            continue


        node_lineno = getattr(
            node,
            "lineno",
            None,
        )


        if not (
            isinstance(
                node_lineno,
                int,
            )

            and

            node_lineno
            <
            cell129_packet_lineno
        ):

            continue


        if isinstance(
            node,
            ast.Assign,
        ):

            targets = (
                node.targets
            )

            value_node = (
                node.value
            )


        else:

            targets = [
                node.target
            ]

            value_node = (
                node.value
            )


        for target in targets:

            for name in cell129_bound_names(
                target
            ):

                cell129_assignment_bindings.setdefault(
                    name,
                    [],
                ).append({
                    "lineno":
                        node_lineno,

                    "binding_type":
                        "ASSIGN",

                    "expression":
                        ast.get_source_segment(
                            cell128_source,
                            value_node,
                        ),

                    "loaded_names":
                        cell129_loaded_names(
                            value_node
                        ),
                })


for name in list(
    cell129_assignment_bindings
):

    cell129_assignment_bindings[
        name
    ] = sorted(
        cell129_assignment_bindings[
            name
        ],
        key=lambda item: item[
            "lineno"
        ],
    )


# ============================================================
# 6) Collect prior FOR-loop bindings
# ============================================================

cell129_for_bindings = {}


if (
    cell128_tree
    is not None

    and

    isinstance(
        cell129_packet_lineno,
        int,
    )
):

    for node in ast.walk(
        cell128_tree
    ):

        if not isinstance(
            node,
            (
                ast.For,
                ast.AsyncFor,
            ),
        ):

            continue


        node_lineno = getattr(
            node,
            "lineno",
            None,
        )


        if not (
            isinstance(
                node_lineno,
                int,
            )

            and

            node_lineno
            <
            cell129_packet_lineno
        ):

            continue


        for name in cell129_bound_names(
            node.target
        ):

            cell129_for_bindings.setdefault(
                name,
                [],
            ).append({
                "lineno":
                    node_lineno,

                "binding_type":
                    "FOR_TARGET",

                "iterator_expression":
                    ast.get_source_segment(
                        cell128_source,
                        node.iter,
                    ),

                "iterator_loaded_names":
                    cell129_loaded_names(
                        node.iter
                    ),
            })


for name in list(
    cell129_for_bindings
):

    cell129_for_bindings[
        name
    ] = sorted(
        cell129_for_bindings[
            name
        ],
        key=lambda item: item[
            "lineno"
        ],
    )


# ============================================================
# 7) Exact pilot_case_id assignment
# ============================================================

cell129_pilot_id_assignments = (
    cell129_assignment_bindings.get(
        "pilot_case_id",
        [],
    )
)


cell129_exact_pilot_id_assignment = (
    cell129_pilot_id_assignments[
        -1
    ]

    if cell129_pilot_id_assignments

    else None
)


cell129_pilot_id_expression = (
    cell129_exact_pilot_id_assignment.get(
        "expression"
    )

    if isinstance(
        cell129_exact_pilot_id_assignment,
        dict,
    )

    else None
)


cell129_pilot_id_loaded_names = (
    cell129_exact_pilot_id_assignment.get(
        "loaded_names",
        [],
    )

    if isinstance(
        cell129_exact_pilot_id_assignment,
        dict,
    )

    else []
)


# ============================================================
# 8) Resolve build_pilot_case_id callable
# ============================================================

cell129_build_pilot_case_id = (
    globals().get(
        "build_pilot_case_id"
    )
)


cell129_build_pilot_case_id_callable = (
    callable(
        cell129_build_pilot_case_id
    )
)


try:

    cell129_build_pilot_case_id_signature = (
        str(
            inspect.signature(
                cell129_build_pilot_case_id
            )
        )

        if cell129_build_pilot_case_id_callable

        else None
    )


except Exception:

    cell129_build_pilot_case_id_signature = None


# ============================================================
# 9) Packet-field loaded Names
# ============================================================

cell129_packet_field_loaded_names = {}


if (
    cell128_packet_node
    is not None
):

    for key_node, value_node in zip(
        cell128_packet_node.keys,
        cell128_packet_node.values,
    ):

        if not (
            isinstance(
                key_node,
                ast.Constant,
            )

            and

            isinstance(
                key_node.value,
                str,
            )
        ):

            continue


        cell129_packet_field_loaded_names[
            key_node.value
        ] = cell129_loaded_names(
            value_node
        )


cell129_all_packet_loaded_names = sorted({
    name

    for names
    in cell129_packet_field_loaded_names.values()

    for name
    in names
})


# ============================================================
# 10) Classify every Packet dependency
# ============================================================

cell129_dependency_resolution = {}


for name in (
    cell129_all_packet_loaded_names
):

    if hasattr(
        builtins,
        name,
    ):

        cell129_dependency_resolution[
            name
        ] = {
            "status":
                "PYTHON_BUILTIN",

            "detail":
                name,
        }

        continue


    assign_matches = (
        cell129_assignment_bindings.get(
            name,
            [],
        )
    )


    if assign_matches:

        cell129_dependency_resolution[
            name
        ] = {
            "status":
                "HISTORICAL_ASSIGNMENT",

            "detail":
                assign_matches[
                    -1
                ],
        }

        continue


    for_matches = (
        cell129_for_bindings.get(
            name,
            [],
        )
    )


    if for_matches:

        cell129_dependency_resolution[
            name
        ] = {
            "status":
                "HISTORICAL_FOR_BINDING",

            "detail":
                for_matches[
                    -1
                ],
        }

        continue


    if name in globals():

        cell129_dependency_resolution[
            name
        ] = {
            "status":
                "RUNTIME_GLOBAL",

            "detail":
                type(
                    globals()[
                        name
                    ]
                ).__name__,
        }

        continue


    cell129_dependency_resolution[
        name
    ] = {
        "status":
            "UNRESOLVED",

        "detail":
            None,
    }


cell129_unresolved_packet_names = [
    name

    for name, resolution
    in cell129_dependency_resolution.items()

    if resolution[
        "status"
    ]
    ==
    "UNRESOLVED"
]


# ============================================================
# 11) Explicit Candidate-local bindings
# ============================================================

CELL129_EXPECTED_CANDIDATE_LOCALS = [
    "selection",
    "case_id",
    "task_draft_id",
    "target_id",
    "pilot_case_id",
]


cell129_candidate_local_resolution = {}


for name in (
    CELL129_EXPECTED_CANDIDATE_LOCALS
):

    assign_matches = (
        cell129_assignment_bindings.get(
            name,
            [],
        )
    )


    for_matches = (
        cell129_for_bindings.get(
            name,
            [],
        )
    )


    if assign_matches:

        cell129_candidate_local_resolution[
            name
        ] = {
            "resolved":
                True,

            "binding_type":
                "ASSIGN",

            "detail":
                assign_matches[
                    -1
                ],
        }


    elif for_matches:

        cell129_candidate_local_resolution[
            name
        ] = {
            "resolved":
                True,

            "binding_type":
                "FOR_TARGET",

            "detail":
                for_matches[
                    -1
                ],
        }


    else:

        cell129_candidate_local_resolution[
            name
        ] = {
            "resolved":
                False,

            "binding_type":
                None,

            "detail":
                None,
        }


cell129_core_candidate_locals_resolved = all(
    item[
        "resolved"
    ]

    for item
    in cell129_candidate_local_resolution.values()
)


# ============================================================
# 12) Confirm expected dependency chain
# ============================================================

cell129_selection_binding = (
    cell129_candidate_local_resolution[
        "selection"
    ]
)


cell129_selection_is_historical_binding = (
    cell129_selection_binding[
        "resolved"
    ]

    and

    cell129_selection_binding[
        "binding_type"
    ]
    in {
        "FOR_TARGET",
        "ASSIGN",
    }
)


cell129_pilot_id_assignment_resolved = (
    isinstance(
        cell129_exact_pilot_id_assignment,
        dict,
    )

    and

    isinstance(
        cell129_pilot_id_expression,
        str,
    )

    and

    "build_pilot_case_id"
    in
    cell129_pilot_id_expression
)


cell129_pilot_id_dependencies_resolved = all(
    (
        name
        ==
        "build_pilot_case_id"

        and

        cell129_build_pilot_case_id_callable
    )

    or

    (
        name
        in
        cell129_candidate_local_resolution

        and

        cell129_candidate_local_resolution[
            name
        ][
            "resolved"
        ]
    )

    for name
    in cell129_pilot_id_loaded_names
)


cell129_dependency_contract_resolved = all([
    cell129_pilot_id_assignment_resolved,
    cell129_build_pilot_case_id_callable,
    cell129_pilot_id_dependencies_resolved,
    cell129_selection_is_historical_binding,
    cell129_core_candidate_locals_resolved,
    len(
        cell129_unresolved_packet_names
    )
    ==
    0,
])


# ============================================================
# 13) Conservative interpretation
# ============================================================

if cell129_dependency_contract_resolved:

    cell129_status_detail = (
        "EXACT_PILOT_ID_AND_PACKET_DEPENDENCY_CONTRACT_RESOLVED"
    )

    cell129_next_action = (
        "SIMULATE_REPLACEMENT_PILOT_PACKET_IN_ISOLATED_LOCAL_ENVIRONMENT"
    )


else:

    cell129_status_detail = (
        "PILOT_PACKET_DEPENDENCY_CONTRACT_INCOMPLETE"
    )

    cell129_next_action = (
        "DO_NOT_SIMULATE_OR_STAGE_REPLACEMENT"
    )


# ============================================================
# 14) Result
# ============================================================

cell129_pilot_packet_dependency_contract = {
    "inspection_version":
        "pilot_packet_dependency_contract_v2",

    "packet_lineno":
        cell129_packet_lineno,

    "pilot_id_assignment_count":
        len(
            cell129_pilot_id_assignments
        ),

    "pilot_id_expression":
        cell129_pilot_id_expression,

    "pilot_id_loaded_names":
        cell129_pilot_id_loaded_names,

    "build_pilot_case_id_callable":
        cell129_build_pilot_case_id_callable,

    "build_pilot_case_id_signature":
        cell129_build_pilot_case_id_signature,

    "packet_field_loaded_names":
        cell129_packet_field_loaded_names,

    "dependency_resolution":
        cell129_dependency_resolution,

    "unresolved_packet_names":
        cell129_unresolved_packet_names,

    "candidate_local_resolution":
        cell129_candidate_local_resolution,

    "selection_is_historical_binding":
        cell129_selection_is_historical_binding,

    "core_candidate_locals_resolved":
        cell129_core_candidate_locals_resolved,

    "pilot_id_dependencies_resolved":
        cell129_pilot_id_dependencies_resolved,

    "dependency_contract_resolved":
        cell129_dependency_contract_resolved,

    "replacement_packet_created":
        False,

    "replacement_staged":
        False,

    "replacement_activated":
        False,

    "queue_advanced":
        False,

    "status":
        cell129_status_detail,

    "next_action":
        cell129_next_action,
}


cell129_pilot_packet_dependency_sha256 = (
    cell129_sha(
        cell129_pilot_packet_dependency_contract
    )
)


# ============================================================
# 15) Freeze protected state AFTER inspection
# ============================================================

cell129_after = {
    "pilots":
        cell129_sha(
            question_drafting_pilot_packets
        ),

    "pilot_index":
        cell129_sha(
            question_drafting_pilot_by_id
        ),

    "queue":
        cell129_sha(
            cell95_remaining_pilot_queue
        ),

    "question":
        cell129_sha(
            cell97_question_draft_record
        ),
}


# ============================================================
# 16) Gates
# ============================================================

cell129_gates = {
    "cell128_is_pass":
        globals().get(
            "cell_128_pass"
        )
        is True,

    "required_objects_exist":
        len(
            cell129_missing_objects
        )
        ==
        0,

    "packet_position_resolved":
        isinstance(
            cell129_packet_lineno,
            int,
        ),

    "pilot_id_assignment_resolved":
        cell129_pilot_id_assignment_resolved,

    "build_pilot_case_id_callable_resolved":
        cell129_build_pilot_case_id_callable,

    "pilot_id_dependencies_resolved":
        cell129_pilot_id_dependencies_resolved,

    "selection_binding_resolved":
        cell129_selection_is_historical_binding,

    "core_candidate_locals_resolved":
        cell129_core_candidate_locals_resolved,

    "python_builtin_list_not_unresolved":
        cell129_dependency_resolution.get(
            "list",
            {},
        ).get(
            "status"
        )
        ==
        "PYTHON_BUILTIN",

    "no_unresolved_packet_names":
        len(
            cell129_unresolved_packet_names
        )
        ==
        0,

    "dependency_contract_resolved":
        cell129_dependency_contract_resolved,

    "replacement_packet_not_created":
        cell129_pilot_packet_dependency_contract[
            "replacement_packet_created"
        ]
        is False,

    "replacement_not_staged":
        cell129_pilot_packet_dependency_contract[
            "replacement_staged"
        ]
        is False,

    "replacement_not_activated":
        cell129_pilot_packet_dependency_contract[
            "replacement_activated"
        ]
        is False,

    "queue_not_advanced":
        cell129_pilot_packet_dependency_contract[
            "queue_advanced"
        ]
        is False,

    "protected_state_byte_stable":
        cell129_before
        ==
        cell129_after,

    "no_artifact_write":
        True,
}


cell_129_pass = all(
    cell129_gates.values()
)


cell129_status = (
    "PILOT_PACKET_DEPENDENCY_CONTRACT_INSPECTED"

    if cell_129_pass

    else
    "PILOT_PACKET_DEPENDENCY_CONTRACT_INSPECTION_FAILED"
)


# ============================================================
# 17) Diagnostics
# ============================================================

print(
    "=== PILOT PACKET DEPENDENCY CONTRACT ==="
)


print(
    "\n=== PILOT CASE ID ==="
)

print(
    "Assignment count:",
    len(
        cell129_pilot_id_assignments
    )
)

print(
    "Resolved expression:",
    cell129_pilot_id_expression
)

print(
    "Loaded names:",
    cell129_pilot_id_loaded_names
)

print(
    "Builder callable:",
    cell129_build_pilot_case_id_callable
)

print(
    "Builder signature:",
    cell129_build_pilot_case_id_signature
)


print(
    "\n=== SELECTION BINDING ==="
)

print(
    cell129_candidate_local_resolution[
        "selection"
    ]
)


print(
    "\n=== CANDIDATE LOCAL BINDINGS ==="
)

for name in (
    CELL129_EXPECTED_CANDIDATE_LOCALS
):

    print(
        name,
        ":",
        cell129_candidate_local_resolution[
            name
        ],
    )


print(
    "\n=== PACKET DEPENDENCY CLASSIFICATION ==="
)

for name in sorted(
    cell129_dependency_resolution
):

    print(
        name,
        ":",
        cell129_dependency_resolution[
            name
        ],
    )


print(
    "\n=== DEPENDENCY RESOLUTION ==="
)

print(
    "Unresolved packet names:",
    cell129_unresolved_packet_names
)

print(
    "Selection binding resolved:",
    cell129_selection_is_historical_binding
)

print(
    "Core candidate locals resolved:",
    cell129_core_candidate_locals_resolved
)

print(
    "Pilot-ID dependencies resolved:",
    cell129_pilot_id_dependencies_resolved
)

print(
    "Dependency contract resolved:",
    cell129_dependency_contract_resolved
)


print(
    "\n=== INTERPRETATION ==="
)

print(
    "Status:",
    cell129_status_detail
)

print(
    "Next action:",
    cell129_next_action
)

print(
    "Inspection SHA256:",
    cell129_pilot_packet_dependency_sha256
)


print(
    "\n=== FAILURES ==="
)

print(
    "Prerequisite failures:",
    cell129_prerequisite_failures
)


print(
    "\nSTATUS:",
    cell129_status
)


print(
    "\n=== CODE CELL 129 GATES ==="
)


for name, passed in (
    cell129_gates.items()
):

    print(
        name,
        ":",
        (
            "PASS"
            if passed
            else "FAIL"
        ),
    )


print(
    "\nFAILED GATES:",
    [
        name

        for name, passed
        in cell129_gates.items()

        if not passed
    ],
)


print(
    "\nCODE CELL 129 GATE:",
    (
        "PASS"
        if cell_129_pass
        else "FAIL"
    ),
)


assert cell_129_pass, (
    "Pilot-packet dependency contract inspection failed. "
    "Do not simulate, stage, or activate a replacement. "
    "Repair this SAME Code Cell 129."
)
```

```python
# Code Cell 130

# ============================================================
# SAME-CELL REPAIR
#
# Use the exact historical selection-loop line resolved by
# Cell 129, execute that loop in an isolated sandbox, reproduce
# all 23 active Pilot packets, then simulate the exclusion
# winner.
#
# No live mutation.
# No staging.
# No activation.
# No queue advancement.
# No artifact write.
# ============================================================

import ast
import copy
import types
import builtins


# ============================================================
# 1) Prerequisites
# ============================================================

cell130_prerequisite_failures = []


if globals().get("cell_129_pass") is not True:
    cell130_prerequisite_failures.append(
        "CELL129_NOT_PASS"
    )


if (
    globals().get("cell129_status")
    !=
    "PILOT_PACKET_DEPENDENCY_CONTRACT_INSPECTED"
):
    cell130_prerequisite_failures.append({
        "reason": "CELL129_STATUS_UNEXPECTED",
        "actual": globals().get("cell129_status"),
    })


CELL130_REQUIRED_OBJECTS = [
    "cell128_tree",
    "cell128_current_packet_keys",
    "cell129_candidate_local_resolution",
    "cell125_exclusion_candidate_summary",
    "cell125_exclusion_result",
    "cell125_exclusion_pool",
    "pilot_selection_rows",
    "question_drafting_pilot_packets",
    "question_drafting_pilot_by_id",
    "target_evidence_records",
    "golden_case_records",
    "task_draft_plan_rows",
    "cell95_remaining_pilot_queue",
    "cell97_question_draft_record",
    "build_pilot_case_id",
]


cell130_missing_objects = [
    name
    for name in CELL130_REQUIRED_OBJECTS
    if name not in globals()
]


if cell130_missing_objects:
    cell130_prerequisite_failures.append({
        "reason": "REQUIRED_RUNTIME_OBJECTS_MISSING",
        "objects": cell130_missing_objects,
    })


# ============================================================
# 2) Helpers
# ============================================================

def cell130_sha(value):

    return hashlib.sha256(
        stable_json_key(value).encode("utf-8")
    ).hexdigest()


def cell130_loaded_names(node):

    return {
        child.id
        for child in ast.walk(node)
        if (
            isinstance(child, ast.Name)
            and
            isinstance(child.ctx, ast.Load)
        )
    }


def cell130_store_names(node):

    return {
        child.id
        for child in ast.walk(node)
        if (
            isinstance(child, ast.Name)
            and
            isinstance(child.ctx, ast.Store)
        )
    }


def cell130_isolated_copy(value):

    if (
        callable(value)
        or isinstance(value, types.ModuleType)
        or isinstance(value, type)
    ):
        return value

    try:
        return copy.deepcopy(value)

    except Exception:
        return value


def cell130_exact_packet_schema(value):

    return (
        isinstance(value, dict)
        and
        sorted(value.keys())
        ==
        cell128_current_packet_keys
    )


# ============================================================
# 3) Freeze protected LIVE state BEFORE
# ============================================================

cell130_before = {
    "pilots":
        cell130_sha(
            question_drafting_pilot_packets
        ),

    "pilot_index":
        cell130_sha(
            question_drafting_pilot_by_id
        ),

    "targets":
        cell130_sha(
            target_evidence_records
        ),

    "cases":
        cell130_sha(
            golden_case_records
        ),

    "tasks":
        cell130_sha(
            task_draft_plan_rows
        ),

    "queue":
        cell130_sha(
            cell95_remaining_pilot_queue
        ),

    "question":
        cell130_sha(
            cell97_question_draft_record
        ),
}


# ============================================================
# 4) Resolve exact historical loop from Cell-129 lineage
# ============================================================

cell130_selection_binding_detail = (
    cell129_candidate_local_resolution[
        "selection"
    ][
        "detail"
    ]
)


cell130_selection_loop_lineno = (
    cell130_selection_binding_detail.get(
        "lineno"
    )
)


cell130_loop_candidates = [
    node
    for node in ast.walk(cell128_tree)
    if (
        isinstance(node, ast.For)
        and
        getattr(node, "lineno", None)
        ==
        cell130_selection_loop_lineno
    )
]


cell130_loop_resolved = (
    len(cell130_loop_candidates)
    ==
    1
)


cell130_loop_node = (
    cell130_loop_candidates[0]
    if cell130_loop_resolved
    else None
)


# ============================================================
# 5) Confirm exact Pilot Packet Dict is inside this loop
# ============================================================

cell130_packet_dict_nodes = []


if cell130_loop_node is not None:

    for node in ast.walk(
        cell130_loop_node
    ):

        if not isinstance(node, ast.Dict):
            continue


        literal_keys = [
            key.value
            for key in node.keys
            if (
                isinstance(key, ast.Constant)
                and
                isinstance(key.value, str)
            )
        ]


        if (
            sorted(literal_keys)
            ==
            cell128_current_packet_keys
        ):
            cell130_packet_dict_nodes.append(
                node
            )


cell130_packet_dict_resolved = (
    len(cell130_packet_dict_nodes)
    ==
    1
)


# ============================================================
# 6) Discover Mutable output names inside historical loop
# ============================================================

cell130_append_targets = set()
cell130_subscript_assignment_targets = set()


if cell130_loop_node is not None:

    for node in ast.walk(
        cell130_loop_node
    ):

        # receiver.append(...)
        if (
            isinstance(node, ast.Call)
            and
            isinstance(node.func, ast.Attribute)
            and
            node.func.attr == "append"
            and
            isinstance(node.func.value, ast.Name)
        ):
            cell130_append_targets.add(
                node.func.value.id
            )


        # mapping[key] = value
        if isinstance(
            node,
            (
                ast.Assign,
                ast.AnnAssign,
            ),
        ):

            targets = (
                node.targets
                if isinstance(node, ast.Assign)
                else [node.target]
            )


            for target in targets:

                if (
                    isinstance(target, ast.Subscript)
                    and
                    isinstance(target.value, ast.Name)
                ):
                    cell130_subscript_assignment_targets.add(
                        target.value.id
                    )


cell130_mutable_output_names = sorted(
    cell130_append_targets
    |
    cell130_subscript_assignment_targets
)


# ============================================================
# 7) Resolve external dependencies
# ============================================================

cell130_loop_loaded_names = (
    cell130_loaded_names(
        cell130_loop_node
    )
    if cell130_loop_node is not None
    else set()
)


cell130_loop_store_names = (
    cell130_store_names(
        cell130_loop_node
    )
    if cell130_loop_node is not None
    else set()
)


cell130_external_dependency_names = sorted(
    cell130_loop_loaded_names
    -
    cell130_loop_store_names
    -
    {
        "selection",
        "pilot_selection_rows",
    }
)


cell130_dependency_resolution = {}


for name in (
    cell130_external_dependency_names
):

    if hasattr(
        builtins,
        name,
    ):
        cell130_dependency_resolution[
            name
        ] = "PYTHON_BUILTIN"


    elif name in globals():
        cell130_dependency_resolution[
            name
        ] = "RUNTIME_GLOBAL"


    else:
        cell130_dependency_resolution[
            name
        ] = "UNRESOLVED"


cell130_unresolved_dependencies = [
    name
    for name, status
    in cell130_dependency_resolution.items()
    if status == "UNRESOLVED"
]


# ============================================================
# 8) Compile exact historical loop with one-row iterator
# ============================================================

cell130_compile_failure = None
cell130_loop_code = None


if (
    cell130_loop_resolved
    and
    cell130_packet_dict_resolved
):

    try:

        loop_copy = copy.deepcopy(
            cell130_loop_node
        )


        loop_copy.iter = ast.Name(
            id="__cell130_selection_rows",
            ctx=ast.Load(),
        )


        module = ast.Module(
            body=[
                loop_copy
            ],
            type_ignores=[],
        )


        ast.fix_missing_locations(
            module
        )


        cell130_loop_code = compile(
            module,
            "<cell130-historical-pilot-loop>",
            "exec",
        )


    except Exception as exc:

        cell130_compile_failure = {
            "reason":
                "HISTORICAL_LOOP_COMPILE_FAILED",

            "error":
                repr(exc),
        }


# ============================================================
# 9) Sandbox execution helper
# ============================================================

def cell130_execute_selection(
    selection_row,
):

    env = {}


    # Copy every resolved external dependency.
    for name, status in (
        cell130_dependency_resolution.items()
    ):

        if status == "RUNTIME_GLOBAL":

            env[name] = (
                cell130_isolated_copy(
                    globals()[name]
                )
            )


    # Isolate frequently used registries defensively.
    for name in [
        "golden_case_records",
        "golden_case_records_by_id",
        "task_draft_plan_rows",
        "task_draft_plans",
        "target_evidence_records",
        "target_evidence_by_case_id",
        "target_evidence_by_task_draft_id",
        "corpus_target_index",
        "case_groundability_precheck",
        "question_drafting_pilot_packets",
        "question_drafting_pilot_by_id",
    ]:

        if name in globals():

            env[name] = (
                cell130_isolated_copy(
                    globals()[name]
                )
            )


    # Reset ONLY names the historical loop itself mutates.
    for name in (
        cell130_append_targets
    ):

        env[name] = []


    for name in (
        cell130_subscript_assignment_targets
    ):

        env[name] = {}


    env[
        "__cell130_selection_rows"
    ] = [
        copy.deepcopy(
            selection_row
        )
    ]


    env[
        "__builtins__"
    ] = dict(
        vars(builtins)
    )


    exec(
        cell130_loop_code,
        env,
        env,
    )


    expected_pilot_id = (
        build_pilot_case_id(
            source_type=
                selection_row[
                    "source_type"
                ],

            stratum=
                selection_row[
                    "stratum"
                ],

            golden_case_id=
                selection_row[
                    "golden_case_id"
                ],

            target_evidence_id=
                selection_row[
                    "target_evidence_id"
                ],
        )
    )


    packet_candidates = []


    # Direct dict variables.
    for value in env.values():

        if cell130_exact_packet_schema(
            value
        ):
            packet_candidates.append(
                value
            )


    # List contents.
    for value in env.values():

        if not isinstance(
            value,
            list,
        ):
            continue


        for child in value:

            if cell130_exact_packet_schema(
                child
            ):
                packet_candidates.append(
                    child
                )


    # Dict values.
    for value in env.values():

        if not isinstance(
            value,
            dict,
        ):
            continue


        for child in value.values():

            if cell130_exact_packet_schema(
                child
            ):
                packet_candidates.append(
                    child
                )


    unique_packets = {}


    for packet in (
        packet_candidates
    ):

        unique_packets[
            cell130_sha(packet)
        ] = packet


    unique_packets = list(
        unique_packets.values()
    )


    identity_matches = [
        packet
        for packet in unique_packets
        if (
            packet.get(
                "pilot_case_id"
            )
            ==
            expected_pilot_id
        )
    ]


    return {
        "expected_pilot_id":
            expected_pilot_id,

        "packet":
            (
                identity_matches[0]
                if len(identity_matches) == 1
                else None
            ),

        "packet_candidate_count":
            len(unique_packets),

        "identity_match_count":
            len(identity_matches),
    }


# ============================================================
# 10) Reproduce all 23 active Pilot packets
# ============================================================

cell130_active_selection_rows = (
    list(
        pilot_selection_rows
    )
)


cell130_reproduction_rows = []
cell130_reproduction_failures = []


if (
    cell130_loop_code is not None
    and
    len(
        cell130_unresolved_dependencies
    )
    ==
    0
):

    for selection_row in (
        cell130_active_selection_rows
    ):

        try:

            result = (
                cell130_execute_selection(
                    selection_row
                )
            )


            persisted_packet = (
                question_drafting_pilot_by_id.get(
                    result[
                        "expected_pilot_id"
                    ]
                )
            )


            exact_match = (
                isinstance(
                    result[
                        "packet"
                    ],
                    dict,
                )
                and
                isinstance(
                    persisted_packet,
                    dict,
                )
                and
                result[
                    "packet"
                ]
                ==
                persisted_packet
            )


            cell130_reproduction_rows.append({
                "pilot_case_id":
                    result[
                        "expected_pilot_id"
                    ],

                "packet_candidate_count":
                    result[
                        "packet_candidate_count"
                    ],

                "identity_match_count":
                    result[
                        "identity_match_count"
                    ],

                "exact_match":
                    exact_match,
            })


        except Exception as exc:

            cell130_reproduction_failures.append({
                "selection":
                    selection_row,

                "reason":
                    "ACTIVE_PILOT_REPRODUCTION_FAILED",

                "error":
                    repr(exc),
            })


cell130_exact_reproduction_count = sum(
    1
    for row
    in cell130_reproduction_rows
    if row["exact_match"]
)


cell130_all_23_reproduced = (
    len(
        cell130_active_selection_rows
    )
    ==
    23
    and
    len(
        cell130_reproduction_rows
    )
    ==
    23
    and
    cell130_exact_reproduction_count
    ==
    23
    and
    len(
        cell130_reproduction_failures
    )
    ==
    0
)


# ============================================================
# 11) Resolve current Selection row
# ============================================================

cell130_current_pilot_id = (
    cell97_question_draft_record[
        "pilot_case_id"
    ]
)


cell130_current_packet = (
    question_drafting_pilot_by_id[
        cell130_current_pilot_id
    ]
)


cell130_current_selection_matches = [
    row
    for row
    in cell130_active_selection_rows
    if (
        row.get("source_type")
        ==
        cell130_current_packet.get(
            "source_type"
        )
        and
        row.get("stratum")
        ==
        cell130_current_packet.get(
            "stratum"
        )
    )
]


cell130_current_selection = (
    cell130_current_selection_matches[0]
    if len(
        cell130_current_selection_matches
    )
    ==
    1
    else None
)


# ============================================================
# 12) Build Candidate Selection with exact current schema
# ============================================================

cell130_candidate_selection = (
    copy.deepcopy(
        cell130_current_selection
    )
    if isinstance(
        cell130_current_selection,
        dict,
    )
    else None
)


cell130_candidate_updates = {}


if isinstance(
    cell130_candidate_selection,
    dict,
):

    candidate = (
        cell125_exclusion_candidate_summary
    )


    proven_values = {
        "golden_case_id":
            candidate[
                "golden_case_id"
            ],

        "task_draft_id":
            candidate[
                "task_draft_id"
            ],

        "target_evidence_id":
            candidate[
                "target_evidence_id"
            ],

        "selected_target_total_tokens":
            candidate[
                "target_total_tokens"
            ],

        "target_total_tokens":
            candidate[
                "target_total_tokens"
            ],

        "distance_to_median":
            candidate[
                "distance_to_median"
            ],

        "selection_tie_break_sha256":
            candidate[
                "stable_tie_break"
            ],

        "stable_tie_break":
            candidate[
                "stable_tie_break"
            ],

        "stratum_target_token_median":
            cell125_exclusion_result[
                "median"
            ],

        "stratum_median_tokens":
            cell125_exclusion_result[
                "median"
            ],

        "stratum_case_count":
            len(
                cell125_exclusion_pool
            ),
    }


    for key, value in (
        proven_values.items()
    ):

        if key in (
            cell130_candidate_selection
        ):

            cell130_candidate_selection[
                key
            ] = value


            cell130_candidate_updates[
                key
            ] = value


cell130_candidate_selection_schema_match = (
    isinstance(
        cell130_candidate_selection,
        dict,
    )
    and
    isinstance(
        cell130_current_selection,
        dict,
    )
    and
    sorted(
        cell130_candidate_selection.keys()
    )
    ==
    sorted(
        cell130_current_selection.keys()
    )
)


cell130_candidate_identity_match = (
    isinstance(
        cell130_candidate_selection,
        dict,
    )
    and
    cell130_candidate_selection.get(
        "golden_case_id"
    )
    ==
    cell125_exclusion_candidate_summary[
        "golden_case_id"
    ]
    and
    cell130_candidate_selection.get(
        "task_draft_id"
    )
    ==
    cell125_exclusion_candidate_summary[
        "task_draft_id"
    ]
    and
    cell130_candidate_selection.get(
        "target_evidence_id"
    )
    ==
    cell125_exclusion_candidate_summary[
        "target_evidence_id"
    ]
)


# ============================================================
# 13) Simulate Candidate
# ============================================================

cell130_candidate_execution_failure = None
cell130_candidate_result = None


if (
    cell130_all_23_reproduced
    and
    cell130_candidate_selection_schema_match
    and
    cell130_candidate_identity_match
):

    try:

        cell130_candidate_result = (
            cell130_execute_selection(
                cell130_candidate_selection
            )
        )


    except Exception as exc:

        cell130_candidate_execution_failure = {
            "reason":
                "CANDIDATE_SIMULATION_FAILED",

            "error":
                repr(exc),
        }


cell130_simulated_packet = (
    cell130_candidate_result[
        "packet"
    ]
    if isinstance(
        cell130_candidate_result,
        dict,
    )
    else None
)


cell130_simulated_pilot_case_id = (
    cell130_candidate_result[
        "expected_pilot_id"
    ]
    if isinstance(
        cell130_candidate_result,
        dict,
    )
    else None
)


# ============================================================
# 14) Candidate validation
# ============================================================

cell130_simulated_schema_match = (
    cell130_exact_packet_schema(
        cell130_simulated_packet
    )
)


cell130_simulated_identity_match = (
    isinstance(
        cell130_simulated_packet,
        dict,
    )
    and
    cell130_simulated_packet.get(
        "pilot_case_id"
    )
    ==
    cell130_simulated_pilot_case_id
    and
    cell130_simulated_packet.get(
        "golden_case_id"
    )
    ==
    cell125_exclusion_candidate_summary[
        "golden_case_id"
    ]
    and
    cell130_simulated_packet.get(
        "task_draft_id"
    )
    ==
    cell125_exclusion_candidate_summary[
        "task_draft_id"
    ]
    and
    cell130_simulated_packet.get(
        "target_evidence_id"
    )
    ==
    cell125_exclusion_candidate_summary[
        "target_evidence_id"
    ]
)


cell130_candidate_target_matches = [
    record
    for record
    in target_evidence_records
    if (
        isinstance(record, dict)
        and
        record.get(
            "target_evidence_id"
        )
        ==
        cell125_exclusion_candidate_summary[
            "target_evidence_id"
        ]
    )
]


cell130_candidate_target_record = (
    cell130_candidate_target_matches[0]
    if len(
        cell130_candidate_target_matches
    )
    ==
    1
    else None
)


cell130_simulated_target_metadata_match = (
    isinstance(
        cell130_simulated_packet,
        dict,
    )
    and
    isinstance(
        cell130_candidate_target_record,
        dict,
    )
    and
    cell130_simulated_packet.get(
        "target_total_tokens"
    )
    ==
    cell130_candidate_target_record.get(
        "target_total_tokens"
    )
    and
    cell130_simulated_packet.get(
        "target_selection_basis"
    )
    ==
    cell130_candidate_target_record.get(
        "target_selection_basis"
    )
)


cell130_existing_pilot_ids = {
    packet[
        "pilot_case_id"
    ]
    for packet
    in question_drafting_pilot_packets
}


cell130_other_active_target_ids = {
    packet[
        "target_evidence_id"
    ]
    for packet
    in question_drafting_pilot_packets
    if (
        packet[
            "pilot_case_id"
        ]
        !=
        cell130_current_pilot_id
    )
}


cell130_pilot_id_collision = (
    cell130_simulated_pilot_case_id
    in
    cell130_existing_pilot_ids
)


cell130_target_collision = (
    cell125_exclusion_candidate_summary[
        "target_evidence_id"
    ]
    in
    cell130_other_active_target_ids
)


# ============================================================
# 15) Simulated lineage
# ============================================================

cell130_simulated_lineage = {
    "lineage_version":
        "pilot_replacement_simulation_lineage_v1",

    "supersedes_pilot_case_id":
        cell130_current_pilot_id,

    "supersedes_target_evidence_id":
        cell125_blocked_target_id,

    "simulated_replacement_pilot_case_id":
        cell130_simulated_pilot_case_id,

    "simulated_replacement_golden_case_id":
        cell125_exclusion_candidate_summary[
            "golden_case_id"
        ],

    "simulated_replacement_task_draft_id":
        cell125_exclusion_candidate_summary[
            "task_draft_id"
        ],

    "simulated_replacement_target_evidence_id":
        cell125_exclusion_candidate_summary[
            "target_evidence_id"
        ],

    "replacement_reason":
        "CURRENT_TARGET_UNSUITABLE_AND_NO_SAFE_LOCAL_WINDOW_ALTERNATIVE",

    "simulation_only":
        True,

    "activation_status":
        "NOT_STARTED",
}


cell130_simulated_lineage_sha256 = (
    cell130_sha(
        cell130_simulated_lineage
    )
)


# ============================================================
# 16) Final decision
# ============================================================

cell130_packet_simulation_valid = all([
    cell130_loop_resolved,
    cell130_packet_dict_resolved,
    cell130_compile_failure is None,
    len(
        cell130_unresolved_dependencies
    )
    ==
    0,
    cell130_all_23_reproduced,
    cell130_candidate_selection_schema_match,
    cell130_candidate_identity_match,
    cell130_candidate_execution_failure
    is None,
    isinstance(
        cell130_simulated_packet,
        dict,
    ),
    cell130_simulated_schema_match,
    cell130_simulated_identity_match,
    cell130_simulated_target_metadata_match,
    not cell130_pilot_id_collision,
    not cell130_target_collision,
])


if cell130_packet_simulation_valid:

    cell130_status_detail = (
        "REPLACEMENT_PILOT_PACKET_SIMULATION_VALID_"
        "NOT_STAGED_NOT_ACTIVATED"
    )

    cell130_next_action = (
        "INSPECT_REPLACEMENT_TRANSACTION_STAGING_BOUNDARY_"
        "BEFORE_ANY_STATE_CHANGE"
    )


else:

    cell130_status_detail = (
        "REPLACEMENT_PILOT_PACKET_SIMULATION_INVALID"
    )

    cell130_next_action = (
        "DO_NOT_STAGE_OR_ACTIVATE_REPLACEMENT"
    )


# ============================================================
# 17) Result
# ============================================================

cell130_replacement_packet_simulation = {
    "simulation_version":
        "replacement_pilot_packet_simulation_v6",

    "selection_loop_lineno":
        cell130_selection_loop_lineno,

    "loop_resolved":
        cell130_loop_resolved,

    "packet_dict_resolved":
        cell130_packet_dict_resolved,

    "mutable_output_names":
        cell130_mutable_output_names,

    "unresolved_dependencies":
        cell130_unresolved_dependencies,

    "active_selection_count":
        len(
            cell130_active_selection_rows
        ),

    "active_reproduction_count":
        len(
            cell130_reproduction_rows
        ),

    "active_exact_reproduction_count":
        cell130_exact_reproduction_count,

    "all_23_reproduced":
        cell130_all_23_reproduced,

    "candidate_selection_schema_match":
        cell130_candidate_selection_schema_match,

    "candidate_selection_updates":
        cell130_candidate_updates,

    "simulated_pilot_case_id":
        cell130_simulated_pilot_case_id,

    "simulated_schema_match":
        cell130_simulated_schema_match,

    "simulated_identity_match":
        cell130_simulated_identity_match,

    "simulated_target_metadata_match":
        cell130_simulated_target_metadata_match,

    "pilot_id_collision":
        cell130_pilot_id_collision,

    "target_collision":
        cell130_target_collision,

    "lineage_sha256":
        cell130_simulated_lineage_sha256,

    "packet_simulation_valid":
        cell130_packet_simulation_valid,

    "replacement_registered":
        False,

    "replacement_staged":
        False,

    "replacement_activated":
        False,

    "queue_advanced":
        False,

    "status":
        cell130_status_detail,

    "next_action":
        cell130_next_action,
}


cell130_replacement_packet_simulation_sha256 = (
    cell130_sha(
        cell130_replacement_packet_simulation
    )
)


# ============================================================
# 18) Freeze protected LIVE state AFTER
# ============================================================

cell130_after = {
    "pilots":
        cell130_sha(
            question_drafting_pilot_packets
        ),

    "pilot_index":
        cell130_sha(
            question_drafting_pilot_by_id
        ),

    "targets":
        cell130_sha(
            target_evidence_records
        ),

    "cases":
        cell130_sha(
            golden_case_records
        ),

    "tasks":
        cell130_sha(
            task_draft_plan_rows
        ),

    "queue":
        cell130_sha(
            cell95_remaining_pilot_queue
        ),

    "question":
        cell130_sha(
            cell97_question_draft_record
        ),
}


# ============================================================
# 19) Gates
# ============================================================

cell130_gates = {
    "cell129_is_pass":
        globals().get(
            "cell_129_pass"
        )
        is True,

    "required_objects_exist":
        len(
            cell130_missing_objects
        )
        ==
        0,

    "historical_selection_loop_resolved":
        cell130_loop_resolved,

    "exact_packet_dict_inside_loop":
        cell130_packet_dict_resolved,

    "historical_loop_compiled":
        (
            cell130_compile_failure
            is None
            and
            cell130_loop_code
            is not None
        ),

    "all_external_dependencies_resolved":
        len(
            cell130_unresolved_dependencies
        )
        ==
        0,

    "active_selection_population_is_23":
        len(
            cell130_active_selection_rows
        )
        ==
        23,

    "all_23_active_packets_reproduced_exactly":
        cell130_all_23_reproduced,

    "candidate_selection_schema_matches":
        cell130_candidate_selection_schema_match,

    "candidate_identity_matches":
        cell130_candidate_identity_match,

    "candidate_execution_succeeded":
        cell130_candidate_execution_failure
        is None,

    "simulated_packet_resolved":
        isinstance(
            cell130_simulated_packet,
            dict,
        ),

    "simulated_packet_schema_matches":
        cell130_simulated_schema_match,

    "simulated_packet_identity_matches":
        cell130_simulated_identity_match,

    "simulated_target_metadata_matches":
        cell130_simulated_target_metadata_match,

    "pilot_id_has_no_active_collision":
        not cell130_pilot_id_collision,

    "target_has_no_other_active_pilot_collision":
        not cell130_target_collision,

    "packet_simulation_valid":
        cell130_packet_simulation_valid,

    "replacement_not_registered":
        cell130_replacement_packet_simulation[
            "replacement_registered"
        ]
        is False,

    "replacement_not_staged":
        cell130_replacement_packet_simulation[
            "replacement_staged"
        ]
        is False,

    "replacement_not_activated":
        cell130_replacement_packet_simulation[
            "replacement_activated"
        ]
        is False,

    "queue_not_advanced":
        cell130_replacement_packet_simulation[
            "queue_advanced"
        ]
        is False,

    "protected_live_state_byte_stable":
        cell130_before
        ==
        cell130_after,

    "no_artifact_write":
        True,
}


cell_130_pass = all(
    cell130_gates.values()
)


cell130_status = (
    "REPLACEMENT_PILOT_PACKET_SIMULATION_INSPECTED"
    if cell_130_pass
    else
    "REPLACEMENT_PILOT_PACKET_SIMULATION_FAILED"
)


# ============================================================
# 20) Diagnostics
# ============================================================

print(
    "=== REPLACEMENT PILOT PACKET — LINEAGE-RESOLVED SANDBOX ==="
)


print(
    "\n=== HISTORICAL LOOP ==="
)

print(
    "Expected line:",
    cell130_selection_loop_lineno
)

print(
    "Candidates:",
    len(
        cell130_loop_candidates
    )
)

print(
    "Resolved:",
    cell130_loop_resolved
)

print(
    "Exact Packet Dict count:",
    len(
        cell130_packet_dict_nodes
    )
)


print(
    "\n=== MUTABLE OUTPUTS ==="
)

print(
    "Append targets:",
    sorted(
        cell130_append_targets
    )
)

print(
    "Subscript-assignment targets:",
    sorted(
        cell130_subscript_assignment_targets
    )
)


print(
    "\n=== DEPENDENCIES ==="
)

print(
    "External:",
    cell130_external_dependency_names
)

print(
    "Resolution:",
    cell130_dependency_resolution
)

print(
    "Unresolved:",
    cell130_unresolved_dependencies
)

print(
    "Compile failure:",
    cell130_compile_failure
)


print(
    "\n=== 23 ACTIVE PILOT REPRODUCTION ==="
)

print(
    "Selection rows:",
    len(
        cell130_active_selection_rows
    )
)

print(
    "Reproduction rows:",
    len(
        cell130_reproduction_rows
    )
)

print(
    "Exact matches:",
    cell130_exact_reproduction_count
)

print(
    "All 23 reproduced:",
    cell130_all_23_reproduced
)


for row in (
    cell130_reproduction_rows
):

    if not row[
        "exact_match"
    ]:

        print(
            "MISMATCH:",
            row
        )


print(
    "Failures:",
    cell130_reproduction_failures
)


print(
    "\n=== CANDIDATE SELECTION ==="
)

print(
    cell130_candidate_selection
)

print(
    "Updates:",
    cell130_candidate_updates
)

print(
    "Schema match:",
    cell130_candidate_selection_schema_match
)

print(
    "Identity match:",
    cell130_candidate_identity_match
)


print(
    "\n=== SIMULATED PACKET ==="
)

print(
    cell130_simulated_packet
)


print(
    "\n=== VALIDATION ==="
)

print(
    "Pilot ID:",
    cell130_simulated_pilot_case_id
)

print(
    "Schema match:",
    cell130_simulated_schema_match
)

print(
    "Identity match:",
    cell130_simulated_identity_match
)

print(
    "Target metadata match:",
    cell130_simulated_target_metadata_match
)

print(
    "Pilot ID collision:",
    cell130_pilot_id_collision
)

print(
    "Target collision:",
    cell130_target_collision
)


print(
    "\n=== INTERPRETATION ==="
)

print(
    "Packet simulation valid:",
    cell130_packet_simulation_valid
)

print(
    "Status:",
    cell130_status_detail
)

print(
    "Next action:",
    cell130_next_action
)

print(
    "Simulation SHA256:",
    cell130_replacement_packet_simulation_sha256
)


print(
    "\n=== CODE CELL 130 GATES ==="
)


for name, passed in (
    cell130_gates.items()
):

    print(
        name,
        ":",
        "PASS"
        if passed
        else "FAIL",
    )


print(
    "\nFAILED GATES:",
    [
        name
        for name, passed
        in cell130_gates.items()
        if not passed
    ],
)


print(
    "\nCODE CELL 130 GATE:",
    "PASS"
    if cell_130_pass
    else "FAIL",
)


assert cell_130_pass, (
    "Lineage-resolved historical Pilot Packet simulation failed. "
    "Do not stage or activate any replacement. "
    "Repair this SAME Code Cell 130."
)
```

```python
# Code Cell 136

# ============================================================
# Inspect exact live-activation payload and permission boundary.
#
# Also resolve downstream Cell-95 state that depends on:
#   - cell95_remaining_pilot_ids
#   - cell95_remaining_pilot_queue
#
# Read-only only.
# No live mutation.
# No queue mutation.
# No activation.
# No artifact write.
# ============================================================

import ast
import copy


# ============================================================
# 1) Prerequisites
# ============================================================

cell136_prerequisite_failures = []


if globals().get("cell_135_pass") is not True:
    cell136_prerequisite_failures.append(
        "CELL135_NOT_PASS"
    )


if (
    globals().get("cell135_status")
    !=
    "QUEUE_SLOT_REPLACEMENT_AND_SUPERSESSION_SIMULATION_INSPECTED"
):
    cell136_prerequisite_failures.append({
        "reason":
            "CELL135_STATUS_UNEXPECTED",

        "actual":
            globals().get(
                "cell135_status"
            ),
    })


CELL136_REQUIRED_OBJECTS = [
    "pilot_selection_rows",
    "question_drafting_pilot_packets",
    "question_drafting_pilot_by_id",
    "cell132_simulated_state",
    "cell135_simulated_queue",
    "cell135_simulated_pilot_supersession_lineage",
    "cell132_old_pilot_id",
    "cell132_new_pilot_id",
    "cell95_remaining_pilot_queue",
    "cell97_question_draft_record",
]


cell136_missing_objects = [
    name
    for name in CELL136_REQUIRED_OBJECTS
    if name not in globals()
]


if cell136_missing_objects:
    cell136_prerequisite_failures.append({
        "reason":
            "REQUIRED_RUNTIME_OBJECTS_MISSING",

        "objects":
            cell136_missing_objects,
    })


# ============================================================
# 2) Helpers
# ============================================================

def cell136_sha(value):

    return hashlib.sha256(
        stable_json_key(value).encode(
            "utf-8"
        )
    ).hexdigest()


def cell136_assignment_names(target):

    if isinstance(
        target,
        ast.Name,
    ):
        return [
            target.id
        ]


    if isinstance(
        target,
        (ast.Tuple, ast.List),
    ):

        names = []

        for child in target.elts:
            names.extend(
                cell136_assignment_names(
                    child
                )
            )

        return names


    if (
        isinstance(
            target,
            ast.Subscript,
        )
        and
        isinstance(
            target.value,
            ast.Name,
        )
    ):
        return [
            target.value.id
        ]


    return []


def cell136_loaded_names(node):

    if node is None:
        return []


    return sorted({
        child.id

        for child in ast.walk(
            node
        )

        if (
            isinstance(
                child,
                ast.Name,
            )
            and
            isinstance(
                child.ctx,
                ast.Load,
            )
        )
    })


def cell136_contains_exact_string(
    value,
    target,
):

    if isinstance(
        value,
        str,
    ):
        return value == target


    if isinstance(
        value,
        dict,
    ):

        return any(
            cell136_contains_exact_string(
                key,
                target,
            )
            or
            cell136_contains_exact_string(
                child,
                target,
            )

            for key, child
            in value.items()
        )


    if isinstance(
        value,
        (list, tuple),
    ):

        return any(
            cell136_contains_exact_string(
                child,
                target,
            )
            for child in value
        )


    return False


# ============================================================
# 3) Freeze protected LIVE state BEFORE inspection
# ============================================================

cell136_before = {
    "pilot_selection_rows":
        cell136_sha(
            pilot_selection_rows
        ),

    "question_drafting_pilot_packets":
        cell136_sha(
            question_drafting_pilot_packets
        ),

    "question_drafting_pilot_by_id":
        cell136_sha(
            question_drafting_pilot_by_id
        ),

    "remaining_queue":
        cell136_sha(
            cell95_remaining_pilot_queue
        ),

    "question":
        cell136_sha(
            cell97_question_draft_record
        ),
}


# ============================================================
# 4) Read notebook history
# ============================================================

cell136_history_failures = []


try:

    cell136_history = list(
        get_ipython()
        .history_manager
        .input_hist_raw
    )


except Exception as exc:

    cell136_history = []

    cell136_history_failures.append({
        "reason":
            "HISTORY_READ_FAILED",

        "error":
            repr(
                exc
            ),
    })


# ============================================================
# 5) Inspect historical activation Cell 74
# ============================================================

cell136_history74_source = (
    cell136_history[74]
    if (
        len(cell136_history) > 74
        and
        isinstance(
            cell136_history[74],
            str,
        )
    )
    else None
)


cell136_history74_tree = None
cell136_history74_parse_failure = None


if isinstance(
    cell136_history74_source,
    str,
):

    try:

        cell136_history74_tree = ast.parse(
            cell136_history74_source
        )


    except Exception as exc:

        cell136_history74_parse_failure = (
            repr(exc)
        )


CELL136_HISTORICAL_ACTIVATION_NAMES = {
    "pilot_selection_rows",
    "target_evidence_by_task_draft_id",
    "cross_stratum_repeated_candidate_hashes",
    "question_drafting_pilot_packets",
    "question_drafting_pilot_by_id",
}


cell136_history74_global_names = []
cell136_history74_mutation_events = []


if cell136_history74_tree is not None:

    cell136_history74_global_names = sorted({
        name

        for node in ast.walk(
            cell136_history74_tree
        )

        if isinstance(
            node,
            ast.Global,
        )

        for name in node.names

        if name
        in
        CELL136_HISTORICAL_ACTIVATION_NAMES
    })


    for node in ast.walk(
        cell136_history74_tree
    ):

        value_node = None
        targets = []


        if isinstance(
            node,
            ast.Assign,
        ):

            value_node = node.value

            for target in node.targets:
                targets.extend(
                    cell136_assignment_names(
                        target
                    )
                )


        elif isinstance(
            node,
            ast.AnnAssign,
        ):

            value_node = node.value

            targets.extend(
                cell136_assignment_names(
                    node.target
                )
            )


        elif isinstance(
            node,
            ast.AugAssign,
        ):

            value_node = node.value

            targets.extend(
                cell136_assignment_names(
                    node.target
                )
            )


        else:
            continue


        for target_name in targets:

            if (
                target_name
                not in
                CELL136_HISTORICAL_ACTIVATION_NAMES
            ):
                continue


            try:

                expression = (
                    ast.unparse(
                        value_node
                    )
                )


            except Exception:

                expression = None


            cell136_history74_mutation_events.append({
                "lineno":
                    getattr(
                        node,
                        "lineno",
                        None,
                    ),

                "target":
                    target_name,

                "expression":
                    expression,

                "loaded_names":
                    cell136_loaded_names(
                        value_node
                    ),
            })


cell136_history74_mutation_events = sorted(
    cell136_history74_mutation_events,
    key=lambda row: (
        row.get("lineno")
        if isinstance(
            row.get("lineno"),
            int,
        )
        else 10**9
    ),
)


cell136_history74_mutation_order = [
    row["target"]
    for row in cell136_history74_mutation_events
]


cell136_history74_has_try = (
    any(
        isinstance(
            node,
            ast.Try,
        )
        for node in ast.walk(
            cell136_history74_tree
        )
    )
    if cell136_history74_tree is not None
    else False
)


cell136_history74_has_assert = (
    any(
        isinstance(
            node,
            ast.Assert,
        )
        for node in ast.walk(
            cell136_history74_tree
        )
    )
    if cell136_history74_tree is not None
    else False
)


# ============================================================
# 6) Inspect historical Cell 95 dependency graph
# ============================================================

cell136_history95_source = (
    cell136_history[95]
    if (
        len(cell136_history) > 95
        and
        isinstance(
            cell136_history[95],
            str,
        )
    )
    else None
)


cell136_history95_tree = None
cell136_history95_parse_failure = None


if isinstance(
    cell136_history95_source,
    str,
):

    try:

        cell136_history95_tree = ast.parse(
            cell136_history95_source
        )


    except Exception as exc:

        cell136_history95_parse_failure = (
            repr(exc)
        )


cell136_cell95_assignments = []


if cell136_history95_tree is not None:

    for node in ast.walk(
        cell136_history95_tree
    ):

        value_node = None
        targets = []


        if isinstance(
            node,
            ast.Assign,
        ):

            value_node = node.value

            for target in node.targets:
                targets.extend(
                    cell136_assignment_names(
                        target
                    )
                )


        elif isinstance(
            node,
            ast.AnnAssign,
        ):

            value_node = node.value

            targets.extend(
                cell136_assignment_names(
                    node.target
                )
            )


        else:
            continue


        for target_name in targets:

            if not target_name.startswith(
                "cell95_"
            ):
                continue


            try:

                expression = (
                    ast.unparse(
                        value_node
                    )
                )


            except Exception:

                expression = None


            cell136_cell95_assignments.append({
                "lineno":
                    getattr(
                        node,
                        "lineno",
                        None,
                    ),

                "target":
                    target_name,

                "expression":
                    expression,

                "loaded_names":
                    cell136_loaded_names(
                        value_node
                    ),
            })


cell136_cell95_assignments = sorted(
    cell136_cell95_assignments,
    key=lambda row: (
        row.get("lineno")
        if isinstance(
            row.get("lineno"),
            int,
        )
        else 10**9
    ),
)


# ============================================================
# 7) Build downstream dependency graph from Remaining IDs /
#    Remaining Queue
# ============================================================

cell136_dependency_roots = {
    "cell95_remaining_pilot_ids",
    "cell95_remaining_pilot_queue",
}


cell136_downstream_cell95_names = set(
    cell136_dependency_roots
)


cell136_dependency_expanded = True


while cell136_dependency_expanded:

    cell136_dependency_expanded = False


    for record in (
        cell136_cell95_assignments
    ):

        target = record[
            "target"
        ]


        loaded = set(
            record[
                "loaded_names"
            ]
        )


        if (
            target
            in
            cell136_downstream_cell95_names
        ):
            continue


        if (
            loaded
            &
            cell136_downstream_cell95_names
        ):

            cell136_downstream_cell95_names.add(
                target
            )

            cell136_dependency_expanded = True


cell136_downstream_cell95_names = sorted(
    cell136_downstream_cell95_names
)


cell136_downstream_assignment_records = [
    record

    for record
    in cell136_cell95_assignments

    if record[
        "target"
    ]
    in
    cell136_downstream_cell95_names
]


# ============================================================
# 8) Inspect runtime companion Remaining-ID list
# ============================================================

cell136_remaining_ids_exists = (
    "cell95_remaining_pilot_ids"
    in globals()
)


cell136_remaining_ids_runtime = (
    globals().get(
        "cell95_remaining_pilot_ids"
    )
)


cell136_remaining_ids_is_list = isinstance(
    cell136_remaining_ids_runtime,
    list,
)


cell136_old_id_indexes = (
    [
        index

        for index, pilot_id
        in enumerate(
            cell136_remaining_ids_runtime
        )

        if pilot_id
        ==
        cell132_old_pilot_id
    ]

    if cell136_remaining_ids_is_list

    else []
)


cell136_new_id_indexes = (
    [
        index

        for index, pilot_id
        in enumerate(
            cell136_remaining_ids_runtime
        )

        if pilot_id
        ==
        cell132_new_pilot_id
    ]

    if cell136_remaining_ids_is_list

    else []
)


cell136_remaining_ids_replaceable_same_slot = (
    cell136_remaining_ids_is_list

    and

    len(
        cell136_old_id_indexes
    )
    ==
    1

    and

    len(
        cell136_new_id_indexes
    )
    ==
    0

    and

    cell136_old_id_indexes[0]
    ==
    cell133_old_queue_index
)


cell136_simulated_remaining_ids = (
    copy.deepcopy(
        cell136_remaining_ids_runtime
    )

    if cell136_remaining_ids_is_list

    else None
)


if cell136_remaining_ids_replaceable_same_slot:

    cell136_simulated_remaining_ids[
        cell136_old_id_indexes[0]
    ] = cell132_new_pilot_id


# ============================================================
# 9) Cross-check simulated IDs against simulated Queue
# ============================================================

cell136_simulated_queue_ids = [
    row.get(
        "pilot_case_id"
    )

    for row
    in cell135_simulated_queue

    if isinstance(
        row,
        dict,
    )
]


cell136_remaining_ids_and_queue_coherent = (
    isinstance(
        cell136_simulated_remaining_ids,
        list,
    )

    and

    cell136_simulated_remaining_ids
    ==
    cell136_simulated_queue_ids
)


# ============================================================
# 10) Inspect runtime Cell-95 objects that still reference old
#     Pilot or already reference new Pilot
# ============================================================

cell136_runtime_cell95_reference_rows = []


for name, value in list(
    globals().items()
):

    if not name.startswith(
        "cell95_"
    ):
        continue


    if not isinstance(
        value,
        (
            dict,
            list,
            tuple,
            str,
        ),
    ):
        continue


    old_hit = (
        cell136_contains_exact_string(
            value,
            cell132_old_pilot_id,
        )
    )


    new_hit = (
        cell136_contains_exact_string(
            value,
            cell132_new_pilot_id,
        )
    )


    if (
        old_hit
        or
        new_hit
    ):

        cell136_runtime_cell95_reference_rows.append({
            "name":
                name,

            "type":
                type(value).__name__,

            "contains_old_pilot":
                old_hit,

            "contains_new_pilot":
                new_hit,

            "downstream_of_queue_or_ids":
                name
                in
                cell136_downstream_cell95_names,
        })


cell136_runtime_cell95_reference_rows = sorted(
    cell136_runtime_cell95_reference_rows,
    key=lambda row: row[
        "name"
    ],
)


# ============================================================
# 11) Identify derived Cell-95 state requiring regeneration
# ============================================================

cell136_direct_activation_queue_names = {
    "cell95_remaining_pilot_ids",
    "cell95_remaining_pilot_queue",
}


cell136_derived_cell95_names = sorted(
    set(
        cell136_downstream_cell95_names
    )
    -
    cell136_direct_activation_queue_names
)


cell136_runtime_derived_cell95_names = [
    name

    for name
    in cell136_derived_cell95_names

    if name in globals()
]


cell136_has_derived_cell95_state = (
    len(
        cell136_runtime_derived_cell95_names
    )
    >
    0
)


# ============================================================
# 12) Build READ-ONLY current activation payload
# ============================================================

cell136_current_activation_payload = {
    "pilot_selection_rows":
        copy.deepcopy(
            cell132_simulated_state[
                "pilot_selection_rows"
            ]
        ),

    "question_drafting_pilot_packets":
        copy.deepcopy(
            cell132_simulated_state[
                "question_drafting_pilot_packets"
            ]
        ),

    "question_drafting_pilot_by_id":
        copy.deepcopy(
            cell132_simulated_state[
                "question_drafting_pilot_by_id"
            ]
        ),

    "cell95_remaining_pilot_ids":
        copy.deepcopy(
            cell136_simulated_remaining_ids
        ),

    "cell95_remaining_pilot_queue":
        copy.deepcopy(
            cell135_simulated_queue
        ),
}


cell136_core_activation_payload_names = sorted(
    cell136_current_activation_payload.keys()
)


cell136_core_payload_coherent = all([
    len(
        cell136_current_activation_payload[
            "pilot_selection_rows"
        ]
    )
    ==
    23,

    len(
        cell136_current_activation_payload[
            "question_drafting_pilot_packets"
        ]
    )
    ==
    23,

    len(
        cell136_current_activation_payload[
            "question_drafting_pilot_by_id"
        ]
    )
    ==
    23,

    isinstance(
        cell136_current_activation_payload[
            "cell95_remaining_pilot_ids"
        ],
        list,
    ),

    len(
        cell136_current_activation_payload[
            "cell95_remaining_pilot_ids"
        ]
    )
    ==
    22,

    len(
        cell136_current_activation_payload[
            "cell95_remaining_pilot_queue"
        ]
    )
    ==
    22,

    cell136_remaining_ids_and_queue_coherent,
])


# ============================================================
# 13) Historical no-op activation registries
# ============================================================

cell136_historical_noop_names = []


for name in [
    "target_evidence_by_task_draft_id",
    "cross_stratum_repeated_candidate_hashes",
]:

    if (
        name
        in
        cell131_activation_mutated_names
    ):

        cell136_historical_noop_names.append(
            name
        )


cell136_current_core_differs_from_history = (
    set(
        cell136_core_activation_payload_names
    )
    !=
    set(
        cell131_activation_mutated_names
    )
)


# ============================================================
# 14) Permission boundary
#
# No explicit permission to mutate live notebook state has been
# recorded for this current replacement transaction.
# ============================================================

cell136_explicit_live_activation_permission = False


cell136_permission_status = (
    "NOT_GRANTED"
)


cell136_permission_requirement = (
    "EXPLICIT_USER_APPROVAL_REQUIRED_BEFORE_LIVE_STATE_MUTATION"
)


cell136_activation_allowed_now = (
    cell136_explicit_live_activation_permission
    is True
)


# ============================================================
# 15) Determine whether payload contract is fully ready
# ============================================================

cell136_payload_contract_resolved = all([
    isinstance(
        cell136_history74_source,
        str,
    ),

    cell136_history74_parse_failure
    is None,

    len(
        cell136_history74_global_names
    )
    >
    0,

    isinstance(
        cell136_history95_source,
        str,
    ),

    cell136_history95_parse_failure
    is None,

    cell136_remaining_ids_replaceable_same_slot,

    cell136_remaining_ids_and_queue_coherent,

    cell136_core_payload_coherent,
])


cell136_activation_prerequisites_complete = (
    cell136_payload_contract_resolved

    and

    not cell136_has_derived_cell95_state
)


# ============================================================
# 16) Conservative interpretation
# ============================================================

if (
    cell136_payload_contract_resolved
    and
    cell136_has_derived_cell95_state
):

    cell136_status_detail = (
        "LIVE_ACTIVATION_CORE_PAYLOAD_RESOLVED_"
        "CELL95_DERIVED_STATE_REGENERATION_REQUIRED_"
        "PERMISSION_NOT_GRANTED"
    )

    cell136_next_action = (
        "SIMULATE_EXACT_REBUILD_OF_CELL95_DERIVED_QUEUE_"
        "AND_WORKFLOW_STATE_BEFORE_ANY_PERMISSIONED_ACTIVATION"
    )


elif (
    cell136_activation_prerequisites_complete
    and
    not cell136_activation_allowed_now
):

    cell136_status_detail = (
        "LIVE_ACTIVATION_PAYLOAD_FULLY_RESOLVED_"
        "PERMISSION_NOT_GRANTED"
    )

    cell136_next_action = (
        "REQUIRE_EXPLICIT_USER_APPROVAL_BEFORE_LIVE_ACTIVATION"
    )


elif (
    cell136_activation_prerequisites_complete
    and
    cell136_activation_allowed_now
):

    cell136_status_detail = (
        "LIVE_ACTIVATION_PAYLOAD_AND_PERMISSION_READY"
    )

    cell136_next_action = (
        "APPLY_ATOMIC_LIVE_ACTIVATION_WITH_ROLLBACK_GUARDS"
    )


else:

    cell136_status_detail = (
        "LIVE_ACTIVATION_PAYLOAD_CONTRACT_INCOMPLETE"
    )

    cell136_next_action = (
        "DO_NOT_MUTATE_LIVE_STATE"
    )


# ============================================================
# 17) Result
# ============================================================

cell136_live_activation_payload_contract = {
    "inspection_version":
        "live_activation_payload_contract_v1",

    "historical_activation_global_names":
        cell136_history74_global_names,

    "historical_activation_mutation_order":
        cell136_history74_mutation_order,

    "historical_activation_has_try":
        cell136_history74_has_try,

    "historical_activation_has_assert":
        cell136_history74_has_assert,

    "core_activation_payload_names":
        cell136_core_activation_payload_names,

    "historical_noop_names_current_transaction":
        sorted(
            cell136_historical_noop_names
        ),

    "remaining_ids_replaceable_same_slot":
        cell136_remaining_ids_replaceable_same_slot,

    "remaining_ids_and_queue_coherent":
        cell136_remaining_ids_and_queue_coherent,

    "cell95_downstream_names":
        cell136_downstream_cell95_names,

    "cell95_derived_runtime_names":
        cell136_runtime_derived_cell95_names,

    "derived_cell95_state_requires_regeneration":
        cell136_has_derived_cell95_state,

    "core_payload_coherent":
        cell136_core_payload_coherent,

    "payload_contract_resolved":
        cell136_payload_contract_resolved,

    "activation_prerequisites_complete":
        cell136_activation_prerequisites_complete,

    "permission_status":
        cell136_permission_status,

    "permission_requirement":
        cell136_permission_requirement,

    "activation_allowed_now":
        cell136_activation_allowed_now,

    "live_state_mutated":
        False,

    "queue_mutated":
        False,

    "replacement_activated":
        False,

    "status":
        cell136_status_detail,

    "next_action":
        cell136_next_action,
}


cell136_live_activation_payload_contract_sha256 = (
    cell136_sha(
        cell136_live_activation_payload_contract
    )
)


# ============================================================
# 18) Freeze protected LIVE state AFTER inspection
# ============================================================

cell136_after = {
    "pilot_selection_rows":
        cell136_sha(
            pilot_selection_rows
        ),

    "question_drafting_pilot_packets":
        cell136_sha(
            question_drafting_pilot_packets
        ),

    "question_drafting_pilot_by_id":
        cell136_sha(
            question_drafting_pilot_by_id
        ),

    "remaining_queue":
        cell136_sha(
            cell95_remaining_pilot_queue
        ),

    "question":
        cell136_sha(
            cell97_question_draft_record
        ),
}


# ============================================================
# 19) Gates
# ============================================================

cell136_gates = {
    "cell135_is_pass":
        globals().get(
            "cell_135_pass"
        )
        is True,

    "required_objects_exist":
        len(
            cell136_missing_objects
        )
        ==
        0,

    "history_read_succeeded":
        len(
            cell136_history_failures
        )
        ==
        0,

    "historical_activation_cell_parsed":
        (
            isinstance(
                cell136_history74_source,
                str,
            )
            and
            cell136_history74_parse_failure
            is None
        ),

    "historical_activation_surface_resolved":
        len(
            cell136_history74_global_names
        )
        >
        0,

    "cell95_source_parsed":
        (
            isinstance(
                cell136_history95_source,
                str,
            )
            and
            cell136_history95_parse_failure
            is None
        ),

    "remaining_ids_runtime_exists":
        cell136_remaining_ids_exists,

    "remaining_ids_replace_same_slot":
        cell136_remaining_ids_replaceable_same_slot,

    "simulated_ids_match_simulated_queue":
        cell136_remaining_ids_and_queue_coherent,

    "core_activation_payload_coherent":
        cell136_core_payload_coherent,

    "payload_contract_resolved":
        cell136_payload_contract_resolved,

    "permission_explicitly_not_granted":
        (
            cell136_permission_status
            ==
            "NOT_GRANTED"
            and
            cell136_activation_allowed_now
            is False
        ),

    "live_state_not_mutated":
        cell136_before
        ==
        cell136_after,

    "queue_not_mutated":
        cell136_live_activation_payload_contract[
            "queue_mutated"
        ]
        is False,

    "replacement_not_activated":
        cell136_live_activation_payload_contract[
            "replacement_activated"
        ]
        is False,

    "no_artifact_write":
        True,
}


cell_136_pass = all(
    cell136_gates.values()
)


cell136_status = (
    "LIVE_ACTIVATION_PAYLOAD_AND_PERMISSION_BOUNDARY_INSPECTED"

    if cell_136_pass

    else
    "LIVE_ACTIVATION_PAYLOAD_AND_PERMISSION_BOUNDARY_FAILED"
)


# ============================================================
# 20) Diagnostics
# ============================================================

print(
    "=== LIVE ACTIVATION PAYLOAD + PERMISSION BOUNDARY ==="
)


print(
    "\n=== HISTORICAL CELL 74 ACTIVATION ==="
)

print(
    "Parse failure:",
    cell136_history74_parse_failure
)

print(
    "Global activation names:",
    cell136_history74_global_names
)

print(
    "Mutation order:",
    cell136_history74_mutation_order
)

print(
    "Has try/rollback structure:",
    cell136_history74_has_try
)

print(
    "Has assert gates:",
    cell136_history74_has_assert
)


print(
    "\n=== CELL 95 DEPENDENCY GRAPH ==="
)

print(
    "Parse failure:",
    cell136_history95_parse_failure
)

print(
    "Remaining-ID runtime exists:",
    cell136_remaining_ids_exists
)

print(
    "Old ID indexes:",
    cell136_old_id_indexes
)

print(
    "New ID indexes:",
    cell136_new_id_indexes
)

print(
    "Replaceable in same slot:",
    cell136_remaining_ids_replaceable_same_slot
)

print(
    "Simulated Remaining IDs:",
    cell136_simulated_remaining_ids
)

print(
    "Simulated Queue IDs:",
    cell136_simulated_queue_ids
)

print(
    "IDs / Queue coherent:",
    cell136_remaining_ids_and_queue_coherent
)


print(
    "\nDownstream Cell95 names:",
    cell136_downstream_cell95_names
)

print(
    "Derived runtime Cell95 names:",
    cell136_runtime_derived_cell95_names
)

print(
    "Derived state requires regeneration:",
    cell136_has_derived_cell95_state
)


print(
    "\n=== CELL 95 OLD/NEW PILOT REFERENCES ==="
)

for row in (
    cell136_runtime_cell95_reference_rows
):

    print(
        row
    )


print(
    "\n=== CURRENT READ-ONLY ACTIVATION PAYLOAD ==="
)

print(
    "Payload names:",
    cell136_core_activation_payload_names
)

print(
    "Core payload coherent:",
    cell136_core_payload_coherent
)

print(
    "Historical no-op names:",
    sorted(
        cell136_historical_noop_names
    )
)


print(
    "\n=== PERMISSION BOUNDARY ==="
)

print(
    "Permission status:",
    cell136_permission_status
)

print(
    "Requirement:",
    cell136_permission_requirement
)

print(
    "Activation allowed now:",
    cell136_activation_allowed_now
)


print(
    "\n=== INTERPRETATION ==="
)

print(
    "Payload contract resolved:",
    cell136_payload_contract_resolved
)

print(
    "Activation prerequisites complete:",
    cell136_activation_prerequisites_complete
)

print(
    "Status:",
    cell136_status_detail
)

print(
    "Next action:",
    cell136_next_action
)

print(
    "Inspection SHA256:",
    cell136_live_activation_payload_contract_sha256
)


print(
    "\n=== FAILURES ==="
)

print(
    "Prerequisite failures:",
    cell136_prerequisite_failures
)

print(
    "History failures:",
    cell136_history_failures
)


print(
    "\nSTATUS:",
    cell136_status
)


print(
    "\n=== CODE CELL 136 GATES ==="
)


for name, passed in (
    cell136_gates.items()
):

    print(
        name,
        ":",
        "PASS"
        if passed
        else "FAIL",
    )


print(
    "\nFAILED GATES:",
    [
        name
        for name, passed
        in cell136_gates.items()
        if not passed
    ],
)


print(
    "\nCODE CELL 136 GATE:",
    "PASS"
    if cell_136_pass
    else "FAIL",
)


assert cell_136_pass, (
    "Live activation payload / permission-boundary inspection failed. "
    "Do not mutate live state. "
    "Repair this SAME Code Cell 136."
)
```

```python
# Code Cell 137

# ============================================================
# SAME-CELL REPAIR
#
# Rebuild ONLY the exact Cell-95 derived state required by the
# replacement transaction.
#
# We do NOT execute the entire historical Cell 95 because its
# old assertions / diagnostics belong to the historical run.
#
# Instead:
#   1) extract exact historical assignments for required state
#   2) execute them sequentially in an isolated namespace
#   3) anchor Remaining IDs / Queue to the already validated
#      simulations from Cells 135–136
#
# No live mutation.
# No activation.
# No queue mutation.
# No artifact write.
# ============================================================

import ast
import copy
import builtins
import types


# ============================================================
# 1) Prerequisites
# ============================================================

cell137_prerequisite_failures = []


if globals().get(
    "cell_136_pass"
) is not True:

    cell137_prerequisite_failures.append(
        "CELL136_NOT_PASS"
    )


if (
    globals().get(
        "cell136_status"
    )
    !=
    "LIVE_ACTIVATION_PAYLOAD_AND_PERMISSION_BOUNDARY_INSPECTED"
):

    cell137_prerequisite_failures.append({
        "reason":
            "CELL136_STATUS_UNEXPECTED",

        "actual":
            globals().get(
                "cell136_status"
            ),
    })


CELL137_REQUIRED_OBJECTS = [
    "cell136_history95_tree",
    "cell136_simulated_remaining_ids",
    "cell135_simulated_queue",
    "cell135_new_queue_row",
    "cell132_simulated_state",
    "cell132_old_pilot_id",
    "cell132_new_pilot_id",
    "cell95_remaining_pilot_ids",
    "cell95_remaining_pilot_queue",
    "question_drafting_pilot_packets",
    "question_drafting_pilot_by_id",
    "pilot_selection_rows",
    "cell97_question_draft_record",
]


cell137_missing_objects = [
    name
    for name in CELL137_REQUIRED_OBJECTS
    if name not in globals()
]


if cell137_missing_objects:

    cell137_prerequisite_failures.append({
        "reason":
            "REQUIRED_RUNTIME_OBJECTS_MISSING",

        "objects":
            cell137_missing_objects,
    })


# ============================================================
# 2) Helpers
# ============================================================

def cell137_sha(value):

    return hashlib.sha256(
        stable_json_key(
            value
        ).encode(
            "utf-8"
        )
    ).hexdigest()


def cell137_target_names(target):

    if isinstance(
        target,
        ast.Name,
    ):
        return [
            target.id
        ]


    if isinstance(
        target,
        (ast.Tuple, ast.List),
    ):

        names = []

        for child in target.elts:

            names.extend(
                cell137_target_names(
                    child
                )
            )

        return names


    return []


def cell137_loaded_names(node):

    return sorted({
        child.id

        for child in ast.walk(
            node
        )

        if (
            isinstance(
                child,
                ast.Name,
            )

            and

            isinstance(
                child.ctx,
                ast.Load,
            )
        )
    })


def cell137_isolated_copy(value):

    if (
        callable(
            value
        )

        or

        isinstance(
            value,
            types.ModuleType,
        )

        or

        isinstance(
            value,
            type,
        )
    ):
        return value


    try:

        return copy.deepcopy(
            value
        )


    except Exception:

        return value


def cell137_contains_exact_string(
    value,
    target,
):

    if isinstance(
        value,
        str,
    ):
        return (
            value
            ==
            target
        )


    if isinstance(
        value,
        dict,
    ):

        return any(
            cell137_contains_exact_string(
                key,
                target,
            )
            or
            cell137_contains_exact_string(
                child,
                target,
            )

            for key, child
            in value.items()
        )


    if isinstance(
        value,
        (
            list,
            tuple,
            set,
        ),
    ):

        return any(
            cell137_contains_exact_string(
                child,
                target,
            )

            for child
            in value
        )


    return False


# ============================================================
# 3) Freeze protected LIVE state BEFORE
# ============================================================

cell137_before = {
    "pilot_selection_rows":
        cell137_sha(
            pilot_selection_rows
        ),

    "pilot_packets":
        cell137_sha(
            question_drafting_pilot_packets
        ),

    "pilot_index":
        cell137_sha(
            question_drafting_pilot_by_id
        ),

    "remaining_ids":
        cell137_sha(
            cell95_remaining_pilot_ids
        ),

    "remaining_queue":
        cell137_sha(
            cell95_remaining_pilot_queue
        ),

    "question":
        cell137_sha(
            cell97_question_draft_record
        ),
}


# ============================================================
# 4) Exact Cell-95 state required by current replacement
# ============================================================

CELL137_REQUIRED_DERIVED_NAMES = [
    "cell95_pilot_ids_in_runtime_order",
    "cell95_remaining_pilot_ids",
    "cell95_remaining_pilot_queue",
    "cell95_remaining_pilot_queue_sha256",
    "cell95_next_pilot_id",
    "cell95_next_pilot_queue_entry",
    "cell95_pilot_workflow_state",
    "cell95_pilot_workflow_state_sha256",
]


# ============================================================
# 5) Extract exact historical assignments
# ============================================================

cell137_assignment_records = []


for node in ast.walk(
    cell136_history95_tree
):

    value_node = None
    targets = []


    if isinstance(
        node,
        ast.Assign,
    ):

        value_node = (
            node.value
        )


        for target in node.targets:

            targets.extend(
                cell137_target_names(
                    target
                )
            )


    elif isinstance(
        node,
        ast.AnnAssign,
    ):

        value_node = (
            node.value
        )


        targets.extend(
            cell137_target_names(
                node.target
            )
        )


    else:

        continue


    for target_name in targets:

        if (
            target_name
            not in
            CELL137_REQUIRED_DERIVED_NAMES
        ):
            continue


        try:

            expression = (
                ast.unparse(
                    value_node
                )
            )


        except Exception:

            expression = None


        cell137_assignment_records.append({
            "lineno":
                getattr(
                    node,
                    "lineno",
                    None,
                ),

            "target":
                target_name,

            "expression":
                expression,

            "loaded_names":
                cell137_loaded_names(
                    value_node
                ),

            "statement_node":
                node,
        })


cell137_assignment_records = sorted(
    cell137_assignment_records,
    key=lambda row: (
        row[
            "lineno"
        ]
        if isinstance(
            row.get(
                "lineno"
            ),
            int,
        )
        else 10**9
    ),
)


# ============================================================
# 6) Resolve one final assignment per required state
# ============================================================

cell137_assignments_by_target = {
    name:
        []
    for name in CELL137_REQUIRED_DERIVED_NAMES
}


for record in (
    cell137_assignment_records
):

    cell137_assignments_by_target[
        record[
            "target"
        ]
    ].append(
        record
    )


cell137_missing_required_assignment_names = [
    name

    for name, records
    in cell137_assignments_by_target.items()

    if len(
        records
    )
    ==
    0
]


cell137_required_assignments_resolved = (
    len(
        cell137_missing_required_assignment_names
    )
    ==
    0
)


# ============================================================
# 7) Build isolated namespace with replacement Pilot state
# ============================================================

cell137_sandbox = {}


# Copy globals needed by the historical assignment expressions.
cell137_all_loaded_names = sorted({
    name

    for record
    in cell137_assignment_records

    for name
    in record[
        "loaded_names"
    ]
})


for name in (
    cell137_all_loaded_names
):

    if hasattr(
        builtins,
        name,
    ):
        continue


    if name in globals():

        cell137_sandbox[
            name
        ] = cell137_isolated_copy(
            globals()[
                name
            ]
        )


# Exact replacement Pilot population from Cell 132.
cell137_sandbox[
    "pilot_selection_rows"
] = copy.deepcopy(
    cell132_simulated_state[
        "pilot_selection_rows"
    ]
)


cell137_sandbox[
    "question_drafting_pilot_packets"
] = copy.deepcopy(
    cell132_simulated_state[
        "question_drafting_pilot_packets"
    ]
)


cell137_sandbox[
    "question_drafting_pilot_by_id"
] = copy.deepcopy(
    cell132_simulated_state[
        "question_drafting_pilot_by_id"
    ]
)


cell137_sandbox[
    "__builtins__"
] = dict(
    vars(
        builtins
    )
)


# ============================================================
# 8) Execute required historical assignments in source order
#
# Remaining IDs / Queue are anchored to the independently
# validated exact simulations from Cells 135–136.
# ============================================================

cell137_execution_failures = []
cell137_executed_assignment_rows = []


for record in (
    cell137_assignment_records
):

    target_name = (
        record[
            "target"
        ]
    )


    # These two structures have already been independently
    # reconstructed using the exact historical Queue constructor.
    if (
        target_name
        ==
        "cell95_remaining_pilot_ids"
    ):

        cell137_sandbox[
            target_name
        ] = copy.deepcopy(
            cell136_simulated_remaining_ids
        )


        cell137_executed_assignment_rows.append({
            "target":
                target_name,

            "lineno":
                record[
                    "lineno"
                ],

            "execution":
                "ANCHORED_FROM_EXACT_CELL136_SIMULATION",
        })


        continue


    if (
        target_name
        ==
        "cell95_remaining_pilot_queue"
    ):

        cell137_sandbox[
            target_name
        ] = copy.deepcopy(
            cell135_simulated_queue
        )


        cell137_executed_assignment_rows.append({
            "target":
                target_name,

            "lineno":
                record[
                    "lineno"
                ],

            "execution":
                "ANCHORED_FROM_EXACT_CELL135_CONSTRUCTOR_SIMULATION",
        })


        continue


    # If the same target has multiple historical assignments,
    # execute all in exact source order; the last result is the
    # historical final value.
    statement = copy.deepcopy(
        record[
            "statement_node"
        ]
    )


    module = ast.Module(
        body=[
            statement
        ],
        type_ignores=[],
    )


    ast.fix_missing_locations(
        module
    )


    try:

        code = compile(
            module,
            "<cell137-cell95-derived-assignment>",
            "exec",
        )


        exec(
            code,
            cell137_sandbox,
            cell137_sandbox,
        )


        cell137_executed_assignment_rows.append({
            "target":
                target_name,

            "lineno":
                record[
                    "lineno"
                ],

            "execution":
                "EXECUTED_EXACT_HISTORICAL_ASSIGNMENT",
        })


    except Exception as exc:

        cell137_execution_failures.append({
            "target":
                target_name,

            "lineno":
                record[
                    "lineno"
                ],

            "expression":
                record[
                    "expression"
                ],

            "error":
                repr(
                    exc
                ),
        })


# ============================================================
# 9) Extract rebuilt state
# ============================================================

cell137_rebuilt_state = {
    name:
        cell137_sandbox.get(
            name
        )

    for name
    in CELL137_REQUIRED_DERIVED_NAMES
}


cell137_all_required_outputs_present = all(
    name in cell137_sandbox

    for name
    in CELL137_REQUIRED_DERIVED_NAMES
)


cell137_rebuilt_runtime_order_ids = (
    cell137_rebuilt_state[
        "cell95_pilot_ids_in_runtime_order"
    ]
)


cell137_rebuilt_remaining_ids = (
    cell137_rebuilt_state[
        "cell95_remaining_pilot_ids"
    ]
)


cell137_rebuilt_queue = (
    cell137_rebuilt_state[
        "cell95_remaining_pilot_queue"
    ]
)


cell137_rebuilt_queue_sha = (
    cell137_rebuilt_state[
        "cell95_remaining_pilot_queue_sha256"
    ]
)


cell137_rebuilt_next_pilot_id = (
    cell137_rebuilt_state[
        "cell95_next_pilot_id"
    ]
)


cell137_rebuilt_next_queue_entry = (
    cell137_rebuilt_state[
        "cell95_next_pilot_queue_entry"
    ]
)


cell137_rebuilt_workflow_state = (
    cell137_rebuilt_state[
        "cell95_pilot_workflow_state"
    ]
)


cell137_rebuilt_workflow_state_sha = (
    cell137_rebuilt_state[
        "cell95_pilot_workflow_state_sha256"
    ]
)


# ============================================================
# 10) Cross-check independent Queue simulations
# ============================================================

cell137_remaining_ids_exact_match = (
    cell137_rebuilt_remaining_ids
    ==
    cell136_simulated_remaining_ids
)


cell137_remaining_queue_exact_match = (
    cell137_rebuilt_queue
    ==
    cell135_simulated_queue
)


cell137_next_pilot_is_replacement = (
    cell137_rebuilt_next_pilot_id
    ==
    cell132_new_pilot_id
)


cell137_next_queue_entry_exact_match = (
    cell137_rebuilt_next_queue_entry
    ==
    cell135_new_queue_row
)


# ============================================================
# 11) Validate full Pilot runtime order
# ============================================================

cell137_runtime_order_valid = (
    isinstance(
        cell137_rebuilt_runtime_order_ids,
        list,
    )

    and

    len(
        cell137_rebuilt_runtime_order_ids
    )
    ==
    23

    and

    cell137_rebuilt_runtime_order_ids.count(
        cell132_old_pilot_id
    )
    ==
    0

    and

    cell137_rebuilt_runtime_order_ids.count(
        cell132_new_pilot_id
    )
    ==
    1
)


# ============================================================
# 12) Validate Workflow State
# ============================================================

cell137_workflow_state_resolved = isinstance(
    cell137_rebuilt_workflow_state,
    dict,
)


cell137_workflow_contains_old_pilot = (
    cell137_contains_exact_string(
        cell137_rebuilt_workflow_state,
        cell132_old_pilot_id,
    )
    if cell137_workflow_state_resolved
    else True
)


cell137_workflow_contains_new_pilot = (
    cell137_contains_exact_string(
        cell137_rebuilt_workflow_state,
        cell132_new_pilot_id,
    )
    if cell137_workflow_state_resolved
    else False
)


cell137_workflow_state_rebuilt = (
    cell137_workflow_state_resolved

    and

    not cell137_workflow_contains_old_pilot

    and

    cell137_workflow_contains_new_pilot
)


# ============================================================
# 13) Verify exact hashes independently
# ============================================================

cell137_expected_queue_sha = (
    hashlib.sha256(
        stable_json_key(
            cell137_rebuilt_queue
        ).encode(
            "utf-8"
        )
    ).hexdigest()
)


cell137_queue_sha_exact = (
    isinstance(
        cell137_rebuilt_queue_sha,
        str,
    )

    and

    cell137_rebuilt_queue_sha
    ==
    cell137_expected_queue_sha
)


cell137_expected_workflow_state_sha = (
    hashlib.sha256(
        stable_json_key(
            cell137_rebuilt_workflow_state
        ).encode(
            "utf-8"
        )
    ).hexdigest()
)


cell137_workflow_sha_exact = (
    isinstance(
        cell137_rebuilt_workflow_state_sha,
        str,
    )

    and

    cell137_rebuilt_workflow_state_sha
    ==
    cell137_expected_workflow_state_sha
)


# ============================================================
# 14) Resolve exact changed Cell-95 state surface
# ============================================================

cell137_state_comparisons = []


for name in (
    CELL137_REQUIRED_DERIVED_NAMES
):

    rebuilt_value = (
        cell137_rebuilt_state[
            name
        ]
    )


    live_exists = (
        name in globals()
    )


    live_value = (
        globals().get(
            name
        )
    )


    changed = (
        True
        if not live_exists
        else (
            rebuilt_value
            !=
            live_value
        )
    )


    cell137_state_comparisons.append({
        "name":
            name,

        "live_exists":
            live_exists,

        "changed":
            bool(
                changed
            ),

        "live_contains_old":
            (
                cell137_contains_exact_string(
                    live_value,
                    cell132_old_pilot_id,
                )
                if live_exists
                else False
            ),

        "rebuilt_contains_old":
            cell137_contains_exact_string(
                rebuilt_value,
                cell132_old_pilot_id,
            ),

        "rebuilt_contains_new":
            cell137_contains_exact_string(
                rebuilt_value,
                cell132_new_pilot_id,
            ),
    })


cell137_changed_state_names = sorted(
    row[
        "name"
    ]

    for row
    in cell137_state_comparisons

    if row[
        "changed"
    ]
)


# ============================================================
# 15) Required activation-time Cell95 mutation surface
# ============================================================

CELL137_REQUIRED_ACTIVATION_CELL95_STATE = {
    "cell95_pilot_ids_in_runtime_order",
    "cell95_remaining_pilot_ids",
    "cell95_remaining_pilot_queue",
    "cell95_remaining_pilot_queue_sha256",
    "cell95_next_pilot_id",
    "cell95_next_pilot_queue_entry",
    "cell95_pilot_workflow_state",
    "cell95_pilot_workflow_state_sha256",
}


cell137_activation_cell95_surface_complete = (
    CELL137_REQUIRED_ACTIVATION_CELL95_STATE
    .issubset(
        set(
            cell137_changed_state_names
        )
    )
)


# ============================================================
# 16) Old Question artifacts remain historical
# ============================================================

cell137_old_question_still_old_pilot = (
    cell97_question_draft_record.get(
        "pilot_case_id"
    )
    ==
    cell132_old_pilot_id
)


cell137_question_not_transferred = (
    cell97_question_draft_record.get(
        "pilot_case_id"
    )
    !=
    cell132_new_pilot_id
)


# ============================================================
# 17) Final rebuild decision
# ============================================================

cell137_cell95_rebuild_valid = all([
    cell137_required_assignments_resolved,
    len(
        cell137_execution_failures
    )
    ==
    0,
    cell137_all_required_outputs_present,
    cell137_remaining_ids_exact_match,
    cell137_remaining_queue_exact_match,
    cell137_next_pilot_is_replacement,
    cell137_next_queue_entry_exact_match,
    cell137_runtime_order_valid,
    cell137_workflow_state_rebuilt,
    cell137_queue_sha_exact,
    cell137_workflow_sha_exact,
    cell137_activation_cell95_surface_complete,
    cell137_old_question_still_old_pilot,
    cell137_question_not_transferred,
])


if cell137_cell95_rebuild_valid:

    cell137_status_detail = (
        "CELL95_DERIVED_STATE_EXACTLY_REBUILT_"
        "ACTIVATION_SURFACE_RESOLVED_"
        "LIVE_STATE_UNCHANGED"
    )


    cell137_next_action = (
        "BUILD_FINAL_ATOMIC_ACTIVATION_PAYLOAD_WITH_ROLLBACK_"
        "GUARDS_WITHOUT_APPLYING"
    )


else:

    cell137_status_detail = (
        "CELL95_DERIVED_STATE_REBUILD_INVALID_OR_INCOMPLETE"
    )


    cell137_next_action = (
        "DO_NOT_ACTIVATE_REPLACEMENT"
    )


# ============================================================
# 18) Result
# ============================================================

cell137_cell95_rebuild_contract = {
    "rebuild_version":
        "cell95_replacement_state_rebuild_v2",

    "required_assignment_names":
        CELL137_REQUIRED_DERIVED_NAMES,

    "missing_required_assignment_names":
        cell137_missing_required_assignment_names,

    "executed_assignment_rows":
        cell137_executed_assignment_rows,

    "execution_failures":
        cell137_execution_failures,

    "remaining_ids_exact_match":
        cell137_remaining_ids_exact_match,

    "remaining_queue_exact_match":
        cell137_remaining_queue_exact_match,

    "next_pilot_id":
        cell137_rebuilt_next_pilot_id,

    "next_pilot_is_replacement":
        cell137_next_pilot_is_replacement,

    "runtime_order_valid":
        cell137_runtime_order_valid,

    "workflow_contains_old_pilot":
        cell137_workflow_contains_old_pilot,

    "workflow_contains_new_pilot":
        cell137_workflow_contains_new_pilot,

    "queue_sha_exact":
        cell137_queue_sha_exact,

    "workflow_sha_exact":
        cell137_workflow_sha_exact,

    "changed_cell95_state_names":
        cell137_changed_state_names,

    "activation_cell95_surface_complete":
        cell137_activation_cell95_surface_complete,

    "cell95_rebuild_valid":
        cell137_cell95_rebuild_valid,

    "permission_status":
        "NOT_GRANTED",

    "live_state_mutated":
        False,

    "queue_mutated":
        False,

    "replacement_activated":
        False,

    "status":
        cell137_status_detail,

    "next_action":
        cell137_next_action,
}


cell137_cell95_rebuild_contract_sha256 = (
    cell137_sha(
        cell137_cell95_rebuild_contract
    )
)


# ============================================================
# 19) Freeze protected LIVE state AFTER
# ============================================================

cell137_after = {
    "pilot_selection_rows":
        cell137_sha(
            pilot_selection_rows
        ),

    "pilot_packets":
        cell137_sha(
            question_drafting_pilot_packets
        ),

    "pilot_index":
        cell137_sha(
            question_drafting_pilot_by_id
        ),

    "remaining_ids":
        cell137_sha(
            cell95_remaining_pilot_ids
        ),

    "remaining_queue":
        cell137_sha(
            cell95_remaining_pilot_queue
        ),

    "question":
        cell137_sha(
            cell97_question_draft_record
        ),
}


# ============================================================
# 20) Gates
# ============================================================

cell137_gates = {
    "cell136_is_pass":
        globals().get(
            "cell_136_pass"
        )
        is True,

    "required_objects_exist":
        len(
            cell137_missing_objects
        )
        ==
        0,

    "required_historical_assignments_resolved":
        cell137_required_assignments_resolved,

    "derived_assignments_executed_without_error":
        len(
            cell137_execution_failures
        )
        ==
        0,

    "all_required_outputs_present":
        cell137_all_required_outputs_present,

    "remaining_ids_match_cell136":
        cell137_remaining_ids_exact_match,

    "remaining_queue_matches_cell135":
        cell137_remaining_queue_exact_match,

    "next_pilot_is_replacement":
        cell137_next_pilot_is_replacement,

    "next_queue_entry_matches_cell135":
        cell137_next_queue_entry_exact_match,

    "runtime_order_rebuilt":
        cell137_runtime_order_valid,

    "workflow_state_rebuilt":
        cell137_workflow_state_rebuilt,

    "queue_sha_recomputed_exactly":
        cell137_queue_sha_exact,

    "workflow_sha_recomputed_exactly":
        cell137_workflow_sha_exact,

    "activation_cell95_surface_complete":
        cell137_activation_cell95_surface_complete,

    "old_question_not_transferred":
        cell137_question_not_transferred,

    "cell95_rebuild_valid":
        cell137_cell95_rebuild_valid,

    "permission_still_not_granted":
        cell137_cell95_rebuild_contract[
            "permission_status"
        ]
        ==
        "NOT_GRANTED",

    "live_state_byte_stable":
        cell137_before
        ==
        cell137_after,

    "queue_not_mutated":
        cell137_cell95_rebuild_contract[
            "queue_mutated"
        ]
        is False,

    "replacement_not_activated":
        cell137_cell95_rebuild_contract[
            "replacement_activated"
        ]
        is False,

    "no_artifact_write":
        True,
}


cell_137_pass = all(
    cell137_gates.values()
)


cell137_status = (
    "CELL95_DERIVED_REPLACEMENT_STATE_REBUILD_INSPECTED"

    if cell_137_pass

    else
    "CELL95_DERIVED_REPLACEMENT_STATE_REBUILD_FAILED"
)


# ============================================================
# 21) Diagnostics
# ============================================================

print(
    "=== CELL 95 — DERIVED STATE REBUILD FROM EXACT HISTORICAL ASSIGNMENTS ==="
)


print(
    "\n=== HISTORICAL ASSIGNMENTS ==="
)

print(
    "Required names:",
    CELL137_REQUIRED_DERIVED_NAMES
)

print(
    "Missing assignment names:",
    cell137_missing_required_assignment_names
)


for record in (
    cell137_assignment_records
):

    print({
        "lineno":
            record[
                "lineno"
            ],

        "target":
            record[
                "target"
            ],

        "loaded_names":
            record[
                "loaded_names"
            ],

        "expression":
            record[
                "expression"
            ],
    })


print(
    "\n=== EXECUTION ==="
)

print(
    "Executed rows:",
    cell137_executed_assignment_rows
)

print(
    "Execution failures:",
    cell137_execution_failures
)


print(
    "\n=== REBUILT REMAINING STATE ==="
)

print(
    "Remaining IDs exact Cell136 match:",
    cell137_remaining_ids_exact_match
)

print(
    "Remaining Queue exact Cell135 match:",
    cell137_remaining_queue_exact_match
)

print(
    "Next Pilot ID:",
    cell137_rebuilt_next_pilot_id
)

print(
    "Next Pilot is replacement:",
    cell137_next_pilot_is_replacement
)

print(
    "Next Queue entry exact Cell135 match:",
    cell137_next_queue_entry_exact_match
)


print(
    "\n=== PILOT RUNTIME ORDER ==="
)

print(
    "Runtime order:",
    cell137_rebuilt_runtime_order_ids
)

print(
    "Runtime order valid:",
    cell137_runtime_order_valid
)


print(
    "\n=== WORKFLOW STATE ==="
)

print(
    "Contains old Pilot:",
    cell137_workflow_contains_old_pilot
)

print(
    "Contains new Pilot:",
    cell137_workflow_contains_new_pilot
)

print(
    "Workflow rebuilt:",
    cell137_workflow_state_rebuilt
)


print(
    "\n=== HASH VALIDATION ==="
)

print(
    "Queue SHA:",
    cell137_rebuilt_queue_sha
)

print(
    "Expected Queue SHA:",
    cell137_expected_queue_sha
)

print(
    "Queue SHA exact:",
    cell137_queue_sha_exact
)

print(
    "Workflow SHA:",
    cell137_rebuilt_workflow_state_sha
)

print(
    "Expected Workflow SHA:",
    cell137_expected_workflow_state_sha
)

print(
    "Workflow SHA exact:",
    cell137_workflow_sha_exact
)


print(
    "\n=== EXACT ACTIVATION-TIME CELL95 SURFACE ==="
)

print(
    "Changed names:",
    cell137_changed_state_names
)

print(
    "Required surface complete:",
    cell137_activation_cell95_surface_complete
)


for row in (
    cell137_state_comparisons
):

    print(
        row
    )


print(
    "\n=== INTERPRETATION ==="
)

print(
    "Cell95 rebuild valid:",
    cell137_cell95_rebuild_valid
)

print(
    "Status:",
    cell137_status_detail
)

print(
    "Next action:",
    cell137_next_action
)

print(
    "Rebuild SHA256:",
    cell137_cell95_rebuild_contract_sha256
)


print(
    "\n=== FAILURES ==="
)

print(
    "Prerequisite failures:",
    cell137_prerequisite_failures
)


print(
    "\nSTATUS:",
    cell137_status
)


print(
    "\n=== CODE CELL 137 GATES ==="
)


for name, passed in (
    cell137_gates.items()
):

    print(
        name,
        ":",
        "PASS"
        if passed
        else "FAIL",
    )


print(
    "\nFAILED GATES:",
    [
        name

        for name, passed
        in cell137_gates.items()

        if not passed
    ],
)


print(
    "\nCODE CELL 137 GATE:",
    "PASS"
    if cell_137_pass
    else "FAIL",
)


assert cell_137_pass, (
    "Exact Cell95 derived-state rebuild failed. "
    "Do not activate replacement. "
    "Repair this SAME Code Cell 137."
)
```

<h3>Repair همان Cell 138 — Trace کردن Rollback Helper واقعی Cell 74</h3>

<p>
Rollback معتبر است فقط اگر مجموع Restoreهای مستقیم Handler و Restoreهای Functionهای فراخوانی‌شده،

<p>

```python
# Code Cell 139

# ============================================================
# ATOMIC LIVE ACTIVATION
#
# User permission has been explicitly granted.
#
# Apply exactly the 11 runtime-object replacements validated
# in Cell 138. On ANY failure after mutation begins, restore
# every protected object from the Cell-138 rollback snapshot.
#
# No file write.
# No Commit / Push / PR.
# No queue advancement beyond same-slot replacement.
# ============================================================

import copy


# ============================================================
# 1) Permission + prerequisites
# ============================================================

cell139_permission_status = (
    "GRANTED_BY_USER"
)


cell139_explicit_live_activation_permission = True


cell139_prerequisite_failures = []


if globals().get(
    "cell_138_pass"
) is not True:

    cell139_prerequisite_failures.append(
        "CELL138_NOT_PASS"
    )


if (
    globals().get(
        "cell138_status"
    )
    !=
    "FINAL_ATOMIC_ACTIVATION_PAYLOAD_AND_PERMISSION_GATE_INSPECTED"
):

    cell139_prerequisite_failures.append({
        "reason":
            "CELL138_STATUS_UNEXPECTED",

        "actual":
            globals().get(
                "cell138_status"
            ),
    })


CELL139_REQUIRED_OBJECTS = [
    "cell138_target_values",
    "cell138_target_sha256",
    "cell138_rollback_snapshot",
    "cell138_rollback_sha256",
    "cell138_live_before_sha256",
    "cell138_apply_order",
    "cell138_rollback_order",
    "cell138_historical_noop_sha256",
    "cell135_simulated_queue",
    "cell132_old_pilot_id",
    "cell132_new_pilot_id",
    "cell97_question_draft_record",
    "target_evidence_by_task_draft_id",
    "cross_stratum_repeated_candidate_hashes",
]


cell139_missing_objects = [
    name
    for name in CELL139_REQUIRED_OBJECTS
    if name not in globals()
]


if cell139_missing_objects:

    cell139_prerequisite_failures.append({
        "reason":
            "REQUIRED_RUNTIME_OBJECTS_MISSING",

        "objects":
            cell139_missing_objects,
    })


# ============================================================
# 2) Helper
# ============================================================

def cell139_sha(value):

    return hashlib.sha256(
        stable_json_key(
            value
        ).encode(
            "utf-8"
        )
    ).hexdigest()


# ============================================================
# 3) Freeze non-mutated protected state
# ============================================================

cell139_question_before_sha256 = (
    cell139_sha(
        cell97_question_draft_record
    )
)


cell139_noop_before_sha256 = {
    "target_evidence_by_task_draft_id":
        cell139_sha(
            target_evidence_by_task_draft_id
        ),

    "cross_stratum_repeated_candidate_hashes":
        cell139_sha(
            cross_stratum_repeated_candidate_hashes
        ),
}


# ============================================================
# 4) Stale-state / drift preflight
# ============================================================

cell139_current_live_sha256 = {
    name:
        cell139_sha(
            globals()[
                name
            ]
        )

    for name
    in cell138_apply_order

    if name in globals()
}


cell139_live_surface_complete = (
    set(
        cell139_current_live_sha256
    )
    ==
    set(
        cell138_apply_order
    )
)


cell139_live_state_matches_cell138_snapshot = (
    cell139_live_surface_complete

    and

    cell139_current_live_sha256
    ==
    cell138_live_before_sha256
)


cell139_target_values_still_match_manifest = (
    {
        name:
            cell139_sha(
                cell138_target_values[
                    name
                ]
            )

        for name
        in cell138_apply_order
    }
    ==
    cell138_target_sha256
)


cell139_rollback_values_still_match_manifest = (
    {
        name:
            cell139_sha(
                cell138_rollback_snapshot[
                    name
                ]
            )

        for name
        in cell138_rollback_order
    }
    ==
    cell138_rollback_sha256
)


cell139_noop_state_matches_cell138 = (
    cell139_noop_before_sha256
    ==
    cell138_historical_noop_sha256
)


cell139_preflight_valid = all([
    len(
        cell139_prerequisite_failures
    )
    ==
    0,

    cell139_explicit_live_activation_permission
    is True,

    cell139_live_surface_complete,

    cell139_live_state_matches_cell138_snapshot,

    cell139_target_values_still_match_manifest,

    cell139_rollback_values_still_match_manifest,

    cell139_noop_state_matches_cell138,
])


# ============================================================
# 5) Activation transaction state
# ============================================================

cell139_applied_names = []

cell139_activation_exception = None

cell139_rollback_attempted = False

cell139_rollback_succeeded = False

cell139_post_activation_gates = {}

cell139_post_activation_valid = False


# ============================================================
# 6) Atomic APPLY
# ============================================================

if cell139_preflight_valid:

    try:

        for name in (
            cell138_apply_order
        ):

            globals()[
                name
            ] = copy.deepcopy(
                cell138_target_values[
                    name
                ]
            )


            cell139_applied_names.append(
                name
            )


        # ====================================================
        # 7) Validate exact target SHA after APPLY
        # ====================================================

        cell139_live_after_apply_sha256 = {
            name:
                cell139_sha(
                    globals()[
                        name
                    ]
                )

            for name
            in cell138_apply_order
        }


        cell139_all_target_sha_exact = (
            cell139_live_after_apply_sha256
            ==
            cell138_target_sha256
        )


        # ====================================================
        # 8) Active Pilot Population invariants
        # ====================================================

        cell139_pilot_count = len(
            question_drafting_pilot_packets
        )


        cell139_pilot_index_count = len(
            question_drafting_pilot_by_id
        )


        cell139_old_pilot_active_occurrences = sum(
            1

            for packet
            in question_drafting_pilot_packets

            if (
                isinstance(
                    packet,
                    dict,
                )

                and

                packet.get(
                    "pilot_case_id"
                )
                ==
                cell132_old_pilot_id
            )
        )


        cell139_new_pilot_active_occurrences = sum(
            1

            for packet
            in question_drafting_pilot_packets

            if (
                isinstance(
                    packet,
                    dict,
                )

                and

                packet.get(
                    "pilot_case_id"
                )
                ==
                cell132_new_pilot_id
            )
        )


        cell139_old_pilot_absent_from_index = (
            cell132_old_pilot_id
            not in
            question_drafting_pilot_by_id
        )


        cell139_new_pilot_present_in_index = (
            cell132_new_pilot_id
            in
            question_drafting_pilot_by_id
        )


        # ====================================================
        # 9) Prove the other 22 Pilots are unchanged
        # ====================================================

        cell139_old_other_pilots = {
            pilot_id:
                packet

            for pilot_id, packet
            in cell138_rollback_snapshot[
                "question_drafting_pilot_by_id"
            ].items()

            if pilot_id
            !=
            cell132_old_pilot_id
        }


        cell139_new_other_pilots = {
            pilot_id:
                packet

            for pilot_id, packet
            in question_drafting_pilot_by_id.items()

            if pilot_id
            !=
            cell132_new_pilot_id
        }


        cell139_other_22_pilots_exact = (
            cell139_old_other_pilots
            ==
            cell139_new_other_pilots
        )


        # ====================================================
        # 10) Queue invariants
        # ====================================================

        cell139_queue_count = len(
            cell95_remaining_pilot_queue
        )


        cell139_queue_exact_target = (
            cell95_remaining_pilot_queue
            ==
            cell135_simulated_queue
        )


        cell139_old_pilot_queue_occurrences = [
            index

            for index, row
            in enumerate(
                cell95_remaining_pilot_queue
            )

            if (
                isinstance(
                    row,
                    dict,
                )

                and

                row.get(
                    "pilot_case_id"
                )
                ==
                cell132_old_pilot_id
            )
        ]


        cell139_new_pilot_queue_occurrences = [
            index

            for index, row
            in enumerate(
                cell95_remaining_pilot_queue
            )

            if (
                isinstance(
                    row,
                    dict,
                )

                and

                row.get(
                    "pilot_case_id"
                )
                ==
                cell132_new_pilot_id
            )
        ]


        cell139_same_queue_slot_preserved = (
            cell139_new_pilot_queue_occurrences
            ==
            [
                0
            ]

            and

            cell95_remaining_pilot_queue[
                0
            ].get(
                "queue_index"
            )
            ==
            1
        )


        cell139_other_21_queue_rows_exact = (
            cell95_remaining_pilot_queue[
                1:
            ]
            ==
            cell138_rollback_snapshot[
                "cell95_remaining_pilot_queue"
            ][
                1:
            ]
        )


        # ====================================================
        # 11) Cell-95 workflow state invariants
        # ====================================================

        cell139_runtime_order_count = len(
            cell95_pilot_ids_in_runtime_order
        )


        cell139_runtime_order_old_occurrences = (
            cell95_pilot_ids_in_runtime_order.count(
                cell132_old_pilot_id
            )
        )


        cell139_runtime_order_new_occurrences = (
            cell95_pilot_ids_in_runtime_order.count(
                cell132_new_pilot_id
            )
        )


        cell139_remaining_ids_count = len(
            cell95_remaining_pilot_ids
        )


        cell139_remaining_ids_old_occurrences = (
            cell95_remaining_pilot_ids.count(
                cell132_old_pilot_id
            )
        )


        cell139_remaining_ids_new_occurrences = (
            cell95_remaining_pilot_ids.count(
                cell132_new_pilot_id
            )
        )


        cell139_next_pilot_is_replacement = (
            cell95_next_pilot_id
            ==
            cell132_new_pilot_id
        )


        cell139_next_queue_entry_is_replacement = (
            isinstance(
                cell95_next_pilot_queue_entry,
                dict,
            )

            and

            cell95_next_pilot_queue_entry.get(
                "pilot_case_id"
            )
            ==
            cell132_new_pilot_id
        )


        cell139_queue_sha_exact = (
            cell95_remaining_pilot_queue_sha256
            ==
            cell137_rebuilt_state[
                "cell95_remaining_pilot_queue_sha256"
            ]
        )


        cell139_workflow_state_exact = (
            cell95_pilot_workflow_state
            ==
            cell137_rebuilt_state[
                "cell95_pilot_workflow_state"
            ]
        )


        cell139_workflow_sha_exact = (
            cell95_pilot_workflow_state_sha256
            ==
            cell137_rebuilt_state[
                "cell95_pilot_workflow_state_sha256"
            ]
        )


        # ====================================================
        # 12) Old Question remains historical
        # ====================================================

        cell139_old_question_still_old_pilot = (
            cell97_question_draft_record.get(
                "pilot_case_id"
            )
            ==
            cell132_old_pilot_id
        )


        cell139_question_not_transferred = (
            cell97_question_draft_record.get(
                "pilot_case_id"
            )
            !=
            cell132_new_pilot_id
        )


        cell139_question_unchanged = (
            cell139_sha(
                cell97_question_draft_record
            )
            ==
            cell139_question_before_sha256
        )


        # ====================================================
        # 13) Historical no-op registries remain unchanged
        # ====================================================

        cell139_noop_after_apply_sha256 = {
            "target_evidence_by_task_draft_id":
                cell139_sha(
                    target_evidence_by_task_draft_id
                ),

            "cross_stratum_repeated_candidate_hashes":
                cell139_sha(
                    cross_stratum_repeated_candidate_hashes
                ),
        }


        cell139_noop_registries_unchanged = (
            cell139_noop_after_apply_sha256
            ==
            cell139_noop_before_sha256
            ==
            cell138_historical_noop_sha256
        )


        # ====================================================
        # 14) Post-activation gates
        # ====================================================

        cell139_post_activation_gates = {
            "all_11_objects_match_target_sha":
                cell139_all_target_sha_exact,

            "pilot_count_is_23":
                cell139_pilot_count
                ==
                23,

            "pilot_index_count_is_23":
                cell139_pilot_index_count
                ==
                23,

            "old_pilot_absent_from_active_packets":
                cell139_old_pilot_active_occurrences
                ==
                0,

            "new_pilot_present_once_in_active_packets":
                cell139_new_pilot_active_occurrences
                ==
                1,

            "old_pilot_absent_from_active_index":
                cell139_old_pilot_absent_from_index,

            "new_pilot_present_in_active_index":
                cell139_new_pilot_present_in_index,

            "other_22_pilots_byte_stable":
                cell139_other_22_pilots_exact,

            "remaining_queue_count_is_22":
                cell139_queue_count
                ==
                22,

            "queue_equals_validated_simulation":
                cell139_queue_exact_target,

            "old_pilot_absent_from_queue":
                len(
                    cell139_old_pilot_queue_occurrences
                )
                ==
                0,

            "new_pilot_present_once_in_same_queue_slot":
                cell139_same_queue_slot_preserved,

            "other_21_queue_rows_byte_stable":
                cell139_other_21_queue_rows_exact,

            "runtime_order_count_is_23":
                cell139_runtime_order_count
                ==
                23,

            "old_pilot_absent_from_runtime_order":
                cell139_runtime_order_old_occurrences
                ==
                0,

            "new_pilot_present_once_in_runtime_order":
                cell139_runtime_order_new_occurrences
                ==
                1,

            "remaining_ids_count_is_22":
                cell139_remaining_ids_count
                ==
                22,

            "old_pilot_absent_from_remaining_ids":
                cell139_remaining_ids_old_occurrences
                ==
                0,

            "new_pilot_present_once_in_remaining_ids":
                cell139_remaining_ids_new_occurrences
                ==
                1,

            "next_pilot_is_replacement":
                cell139_next_pilot_is_replacement,

            "next_queue_entry_is_replacement":
                cell139_next_queue_entry_is_replacement,

            "queue_sha_exact":
                cell139_queue_sha_exact,

            "workflow_state_exact":
                cell139_workflow_state_exact,

            "workflow_sha_exact":
                cell139_workflow_sha_exact,

            "old_question_remains_historical":
                cell139_old_question_still_old_pilot,

            "old_question_not_transferred":
                cell139_question_not_transferred,

            "old_question_byte_stable":
                cell139_question_unchanged,

            "historical_noop_registries_unchanged":
                cell139_noop_registries_unchanged,
        }


        cell139_post_activation_valid = all(
            cell139_post_activation_gates.values()
        )


        if not cell139_post_activation_valid:

            raise AssertionError(
                "POST_ACTIVATION_INVARIANT_FAILURE"
            )


    except Exception as exc:

        cell139_activation_exception = repr(
            exc
        )


        # ====================================================
        # 15) Automatic rollback
        # ====================================================

        cell139_rollback_attempted = True


        try:

            for name in (
                cell138_rollback_order
            ):

                globals()[
                    name
                ] = copy.deepcopy(
                    cell138_rollback_snapshot[
                        name
                    ]
                )


            cell139_after_rollback_sha256 = {
                name:
                    cell139_sha(
                        globals()[
                            name
                        ]
                    )

                for name
                in cell138_rollback_order
            }


            cell139_rollback_succeeded = (
                cell139_after_rollback_sha256
                ==
                cell138_rollback_sha256
            )


        except Exception as rollback_exc:

            cell139_rollback_succeeded = False


            cell139_activation_exception = {
                "activation_error":
                    cell139_activation_exception,

                "rollback_error":
                    repr(
                        rollback_exc
                    ),
            }


else:

    cell139_activation_exception = (
        "PREFLIGHT_FAILED_NO_MUTATION_ATTEMPTED"
    )


# ============================================================
# 16) Final transaction state
# ============================================================

cell139_activation_applied = (
    cell139_preflight_valid
    and
    cell139_post_activation_valid
    and
    cell139_activation_exception
    is None
    and
    not cell139_rollback_attempted
)


cell139_live_state_mutated_and_retained = (
    cell139_activation_applied
)


cell139_queue_replacement_applied = (
    cell139_activation_applied
    and
    cell95_next_pilot_id
    ==
    cell132_new_pilot_id
)


# ============================================================
# 17) Activation record
# ============================================================

cell139_activation_record = {
    "activation_version":
        "pilot_replacement_atomic_live_activation_v1",

    "permission_status":
        cell139_permission_status,

    "old_pilot_case_id":
        cell132_old_pilot_id,

    "replacement_pilot_case_id":
        cell132_new_pilot_id,

    "preflight_valid":
        cell139_preflight_valid,

    "applied_object_count":
        len(
            cell139_applied_names
        ),

    "applied_names":
        cell139_applied_names,

    "post_activation_valid":
        cell139_post_activation_valid,

    "rollback_attempted":
        cell139_rollback_attempted,

    "rollback_succeeded":
        cell139_rollback_succeeded,

    "activation_exception":
        cell139_activation_exception,

    "activation_applied":
        cell139_activation_applied,

    "queue_replacement_applied":
        cell139_queue_replacement_applied,

    "file_write":
        False,

    "commit":
        False,

    "push":
        False,

    "pr":
        False,

    "queue_advanced_beyond_same_slot_replacement":
        False,
}


cell139_activation_record_sha256 = (
    cell139_sha(
        cell139_activation_record
    )
)


# ============================================================
# 18) Final Gates
# ============================================================

cell139_gates = {
    "cell138_is_pass":
        globals().get(
            "cell_138_pass"
        )
        is True,

    "explicit_permission_granted":
        cell139_explicit_live_activation_permission,

    "required_objects_exist":
        len(
            cell139_missing_objects
        )
        ==
        0,

    "live_state_matches_cell138_pre_snapshot":
        cell139_live_state_matches_cell138_snapshot,

    "target_manifest_valid":
        cell139_target_values_still_match_manifest,

    "rollback_manifest_valid":
        cell139_rollback_values_still_match_manifest,

    "noop_prestate_valid":
        cell139_noop_state_matches_cell138,

    "preflight_valid":
        cell139_preflight_valid,

    "all_objects_applied":
        cell139_applied_names
        ==
        cell138_apply_order,

    "post_activation_valid":
        cell139_post_activation_valid,

    "rollback_not_needed":
        cell139_rollback_attempted
        is False,

    "activation_applied":
        cell139_activation_applied,

    "queue_replacement_applied":
        cell139_queue_replacement_applied,

    "no_file_write":
        cell139_activation_record[
            "file_write"
        ]
        is False,

    "no_commit_push_pr":
        (
            cell139_activation_record[
                "commit"
            ]
            is False
            and
            cell139_activation_record[
                "push"
            ]
            is False
            and
            cell139_activation_record[
                "pr"
            ]
            is False
        ),

    "no_queue_advance_beyond_replacement":
        cell139_activation_record[
            "queue_advanced_beyond_same_slot_replacement"
        ]
        is False,
}


cell_139_pass = all(
    cell139_gates.values()
)


cell139_status = (
    "REPLACEMENT_PILOT_ATOMICALLY_ACTIVATED_LIVE_STATE_VALID"

    if cell_139_pass

    else
    (
        "REPLACEMENT_ACTIVATION_FAILED_AND_ROLLED_BACK"

        if cell139_rollback_attempted
        and
        cell139_rollback_succeeded

        else
        "REPLACEMENT_ACTIVATION_FAILED"
    )
)


# ============================================================
# 19) Diagnostics
# ============================================================

print(
    "=== ATOMIC LIVE REPLACEMENT ACTIVATION ==="
)


print(
    "\n=== PERMISSION ==="
)

print(
    "Permission:",
    cell139_permission_status
)


print(
    "\n=== PREFLIGHT ==="
)

print(
    "Prerequisite failures:",
    cell139_prerequisite_failures
)

print(
    "Live state matches Cell138 snapshot:",
    cell139_live_state_matches_cell138_snapshot
)

print(
    "Target manifest valid:",
    cell139_target_values_still_match_manifest
)

print(
    "Rollback manifest valid:",
    cell139_rollback_values_still_match_manifest
)

print(
    "No-op state valid:",
    cell139_noop_state_matches_cell138
)

print(
    "Preflight valid:",
    cell139_preflight_valid
)


print(
    "\n=== APPLY ==="
)

print(
    "Applied names:",
    cell139_applied_names
)

print(
    "Activation exception:",
    cell139_activation_exception
)


print(
    "\n=== POST-ACTIVATION GATES ==="
)

for name, passed in (
    cell139_post_activation_gates.items()
):

    print(
        name,
        ":",
        "PASS"
        if passed
        else "FAIL",
    )


print(
    "\n=== ROLLBACK ==="
)

print(
    "Rollback attempted:",
    cell139_rollback_attempted
)

print(
    "Rollback succeeded:",
    cell139_rollback_succeeded
)


print(
    "\n=== FINAL ACTIVE STATE ==="
)

if cell139_activation_applied:

    print(
        "Pilot count:",
        len(
            question_drafting_pilot_packets
        )
    )

    print(
        "Remaining queue count:",
        len(
            cell95_remaining_pilot_queue
        )
    )

    print(
        "Active next Pilot:",
        cell95_next_pilot_id
    )

    print(
        "Old Pilot active:",
        cell132_old_pilot_id
        in
        question_drafting_pilot_by_id
    )

    print(
        "Replacement Pilot active:",
        cell132_new_pilot_id
        in
        question_drafting_pilot_by_id
    )

    print(
        "Old Question still historical:",
        cell97_question_draft_record.get(
            "pilot_case_id"
        )
        ==
        cell132_old_pilot_id
    )


print(
    "\nActivation Record SHA256:",
    cell139_activation_record_sha256
)


print(
    "\nSTATUS:",
    cell139_status
)


print(
    "\n=== CODE CELL 139 GATES ==="
)


for name, passed in (
    cell139_gates.items()
):

    print(
        name,
        ":",
        "PASS"
        if passed
        else "FAIL",
    )


print(
    "\nFAILED GATES:",
    [
        name

        for name, passed
        in cell139_gates.items()

        if not passed
    ],
)


print(
    "\nCODE CELL 139 GATE:",
    "PASS"
    if cell_139_pass
    else "FAIL",
)


assert cell_139_pass, (
    "Atomic replacement activation did not complete successfully. "
    "If rollback_succeeded=True, live state was restored exactly. "
    "Repair this SAME Code Cell 139 before proceeding."
)
```
