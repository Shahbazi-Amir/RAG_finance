## اقدام ۲ — شناسایی نقشه‌های مرجع فاز دوم

**شناسه:** `P2AUD-C002`

### هدف

پیش از مقایسهٔ نقشه‌ها، هویت و نسخهٔ فایل‌های مرجع را از مخزن محلی شناسایی کنیم.

### ورودی‌ها

۱. نقشهٔ ماژولار فاز دوم: `RAG_finance_phase2_modular_notebooks_execution_plan.md`

۲. نقشهٔ عملیاتی فاز دوم: `RAG_finance_phase2_operational_roadmap.md`

۳. نقشهٔ کلان: `RAG_finance_master_roadmap.md`

### کنترل‌ها

- شناسایی ریشهٔ واقعی مخزن از مسیر نوت‌بوک.
- بررسی وجود و خوانایی هر سه فایل.
- محاسبهٔ SHA-256 و Git Blob SHA.
- مقایسهٔ نسخهٔ محلی با نسخهٔ مرجع قبلاً مشاهده‌شده.
- ثبت شاخه و HEAD محلی در صورت دسترسی.
- حفظ متن کامل فایل‌ها برای ممیزی مراحل بعد.

### شرط پایان

هر سه نقشه خوانده شده باشند و وضعیت تطابق نسخه‌ها صریح ثبت شود. اختلاف نسخه نباید پنهان یا خودکار اصلاح شود.

**محدودیت:** هیچ فایل، Artifact یا گزارش جدیدی در این اقدام نوشته نمی‌شود؛ هیچ Pipeline یا API اجرا نمی‌شود و هیچ تغییری در Git انجام نمی‌گیرد.

</div>


````python
# P2AUD-C002 — Read-only discovery and fingerprinting of Phase 2 roadmaps
# Action 2 | NO API | NO FILE WRITES | NO GIT WRITES

assert P2AUD_BOOTSTRAP["status"] == "P2AUD_C001_BOOTSTRAP_READY"

P2AUD_ROADMAP_SPECS = (
    {
        "id": "PHASE2_MODULAR",
        "path": "data/Road_Maps/RAG_finance_phase2_modular_notebooks_execution_plan.md",
        "reference_blob_sha": "e3ec855f54922b7e18d7541bf1aadb39d5c759d3",
    },
    {
        "id": "PHASE2_OPERATIONAL",
        "path": "data/Road_Maps/RAG_finance_phase2_operational_roadmap.md",
        "reference_blob_sha": "a8215a10695e9bed8ab57d78cac8cdc059e486d3",
    },
    {
        "id": "MASTER_ROADMAP",
        "path": "data/Road_Maps/RAG_finance_master_roadmap.md",
        "reference_blob_sha": "c239b7081e88e1f9d806a3e8095f6482a20cbf86",
    },
)

# Normally leave this as None.
# If the notebook is outside the repository, set the local root explicitly.
P2AUD_REPO_ROOT_OVERRIDE = None

if P2AUD_REPO_ROOT_OVERRIDE is None:
    candidate_roots = (
        Path.cwd(),
        *Path.cwd().parents,
    )
else:
    candidate_roots = (
        Path(P2AUD_REPO_ROOT_OVERRIDE).expanduser().resolve(),
    )

P2AUD_REPO_ROOT = None

for candidate in candidate_roots:
    if all(
        (candidate / spec["path"]).is_file()
        for spec in P2AUD_ROADMAP_SPECS
    ):
        P2AUD_REPO_ROOT = candidate.resolve()
        break

if P2AUD_REPO_ROOT is None:
    raise FileNotFoundError(
        "P2AUD_C002_REPOSITORY_ROOT_NOT_FOUND: "
        "The three required roadmap files were not found together. "
        "Check your notebook location or set "
        "P2AUD_REPO_ROOT_OVERRIDE to the local repository root."
    )

P2AUD_ROADMAP_TEXTS = {}
P2AUD_ROADMAP_INDEX = {}

for spec in P2AUD_ROADMAP_SPECS:
    source_path = P2AUD_REPO_ROOT / spec["path"]

    raw = source_path.read_bytes()
    content = raw.decode("utf-8-sig")

    if not content.strip():
        raise ValueError(
            f"Empty roadmap: {spec['id']}"
        )

    source_lines = content.splitlines()

    headings = [
        {
            "line": number,
            "text": line.strip(),
        }
        for number, line in enumerate(source_lines, start=1)
        if re.match(r"^#{1,6}\s+", line)
    ]

    if not headings:
        raise ValueError(
            f"No Markdown headings found in {spec['id']}"
        )

    sha256 = hashlib.sha256(raw).hexdigest()

    git_blob_header = (
        b"blob " + str(len(raw)).encode("ascii") + b"\0"
    )

    git_blob_sha = hashlib.sha1(
        git_blob_header + raw
    ).hexdigest()

    version_status = (
        "MATCH_REFERENCE"
        if git_blob_sha == spec["reference_blob_sha"]
        else "DIFFERS_FROM_REFERENCE"
    )

    P2AUD_ROADMAP_TEXTS[spec["id"]] = content

    P2AUD_ROADMAP_INDEX[spec["id"]] = {
        "relative_path": spec["path"],
        "size_bytes": len(raw),
        "line_count": len(source_lines),
        "heading_count": len(headings),
        "sha256": sha256,
        "git_blob_sha": git_blob_sha,
        "reference_blob_sha": spec["reference_blob_sha"],
        "version_status": version_status,
        "headings": headings,
    }

def p2aud_read_git_value(*arguments):
    try:
        result = subprocess.run(
            ["git", "-C", str(P2AUD_REPO_ROOT), *arguments],
            capture_output=True,
            text=True,
            check=False,
            timeout=8,
        )
        if result.returncode == 0:
            return result.stdout.strip() or None
    except (OSError, subprocess.TimeoutExpired):
        pass
    return None

P2AUD_LOCAL_BRANCH = p2aud_read_git_value(
    "branch", "--show-current"
)

P2AUD_LOCAL_HEAD = p2aud_read_git_value(
    "rev-parse", "HEAD"
)

version_counts = Counter(
    item["version_status"]
    for item in P2AUD_ROADMAP_INDEX.values()
)

all_reference_versions_match = (
    version_counts["MATCH_REFERENCE"]
    == len(P2AUD_ROADMAP_SPECS)
)

P2AUD_C002_REPORT = {
    "schema_version": "P2AUD_ROADMAP_SOURCE_INVENTORY_V1",
    "cell_id": "P2AUD-C002",
    "action": 2,
    "status": (
        "P2AUD_C002_ROADMAP_SOURCES_VERIFIED"
        if all_reference_versions_match
        else "P2AUD_C002_VERSION_DIFFERENCES_REQUIRE_REVIEW"
    ),
    "repository_root_name": P2AUD_REPO_ROOT.name,
    "local_branch": P2AUD_LOCAL_BRANCH,
    "local_head": P2AUD_LOCAL_HEAD,
    "branch_matches_expected": (
        P2AUD_LOCAL_BRANCH == "rag-phase2"
    ),
    "roadmaps_loaded": len(P2AUD_ROADMAP_INDEX),
    "version_counts": dict(version_counts),
    "roadmaps": {
        source_id: {
            key: value
            for key, value in item.items()
            if key != "headings"
        }
        for source_id, item in P2AUD_ROADMAP_INDEX.items()
    },
    "requirements_evaluated": 0,
    "files_written": 0,
    "new_api_calls": 0,
    "git_write_operations": 0,
    "next_cell": "P2AUD-C003",
}

assert P2AUD_C002_REPORT["roadmaps_loaded"] == 3
assert len(P2AUD_ROADMAP_TEXTS) == 3

print(
    json.dumps(
        P2AUD_C002_REPORT,
        ensure_ascii=False,
        indent=2,
    )
)
````

**خروجی:** گزارش ساختاری C003 و داده‌های استخراج‌شده برای C004.

````python
# P2AUD-C003 — Extract Roadmap Structure and Section Boundaries
# Action 3 | READ_ONLY | NO_API | NO_GIT_WRITES

assert P2AUD_C002_REPORT["status"] == (
    "P2AUD_C002_ROADMAP_SOURCES_VERIFIED"
)

assert len(P2AUD_ROADMAP_TEXTS) == 3

P2AUD_C003_HEADINGS = {}
P2AUD_C003_UNITS = {
    "notebooks": {},
    "operational_packages": {},
    "master_sections": {},
}

def p2aud_extract_headings(source_id):
    lines = P2AUD_ROADMAP_TEXTS[source_id].splitlines()
    headings = []

    for number, line in enumerate(lines, start=1):
        match = re.match(
            r"^(#{1,6})[ \t]+(.+?)\s*$",
            line,
        )

        if match:
            headings.append({
                "line": number,
                "level": len(match.group(1)),
                "title": match.group(2).strip(),
            })

    return lines, headings


def p2aud_extract_unit(lines, headings, heading):
    start = heading["line"]
    end = len(lines)

    for next_heading in headings:
        if (
            next_heading["line"] > start
            and next_heading["level"] <= heading["level"]
        ):
            end = next_heading["line"] - 1
            break

    section_lines = lines[start - 1:end]

    subsections = [
        {
            "line": item["line"],
            "title": item["title"],
        }
        for item in headings
        if (
            start < item["line"] <= end
            and item["level"] > heading["level"]
        )
    ]

    candidate_items = [
        {
            "line": number,
            "text": text.strip(),
        }
        for number, text in enumerate(
            section_lines,
            start=start,
        )
        if re.match(
            r"^\s*(?:[-*]|\d+[.)])\s+\S",
            text,
        )
    ]

    raw_section = "\n".join(section_lines)

    return {
        "title": heading["title"],
        "start_line": start,
        "end_line": end,
        "line_count": len(section_lines),
        "subsections": subsections,
        "candidate_items": candidate_items,
        "section_sha256": hashlib.sha256(
            raw_section.encode("utf-8")
        ).hexdigest(),
        "raw_text": raw_section,
    }


for source_id in (
    "PHASE2_MODULAR",
    "PHASE2_OPERATIONAL",
    "MASTER_ROADMAP",
):
    lines, headings = p2aud_extract_headings(
        source_id
    )

    expected_count = P2AUD_ROADMAP_INDEX[
        source_id
    ]["heading_count"]

    assert len(headings) == expected_count, (
        f"Heading count changed: {source_id}"
    )

    P2AUD_C003_HEADINGS[source_id] = headings

    for heading in headings:
        if heading["level"] != 2:
            continue

        title = heading["title"]

        if source_id == "PHASE2_MODULAR":
            match = re.match(
                r"^Notebook\s+(\d{2})\s+[—-]",
                title,
            )

            if match:
                unit_id = "NB" + match.group(1)

                if unit_id in P2AUD_C003_UNITS[
                    "notebooks"
                ]:
                    raise ValueError(
                        f"Duplicate notebook: {unit_id}"
                    )

                P2AUD_C003_UNITS[
                    "notebooks"
                ][unit_id] = p2aud_extract_unit(
                    lines, headings, heading
                )

        elif source_id == "PHASE2_OPERATIONAL":
            match = re.search(
                r"\bP2-(\d{2})\b",
                title,
            )

            if match:
                unit_id = "P2-" + match.group(1)

                if unit_id in P2AUD_C003_UNITS[
                    "operational_packages"
                ]:
                    raise ValueError(
                        f"Duplicate package: {unit_id}"
                    )

                P2AUD_C003_UNITS[
                    "operational_packages"
                ][unit_id] = p2aud_extract_unit(
                    lines, headings, heading
                )

        elif source_id == "MASTER_ROADMAP":
            if re.match(
                r"^[۶6][.٫]\s*فاز\s*[۲2]",
                title,
            ):
                P2AUD_C003_UNITS[
                    "master_sections"
                ]["PHASE2_SCOPE"] = (
                    p2aud_extract_unit(
                        lines, headings, heading
                    )
                )

            elif (
                "قرارداد تحویل بین فازها" in title
            ):
                P2AUD_C003_UNITS[
                    "master_sections"
                ]["PHASE_HANDOFF"] = (
                    p2aud_extract_unit(
                        lines, headings, heading
                    )
                )


expected_notebooks = {
    f"NB{i:02d}" for i in range(24)
}

expected_packages = {
    f"P2-{i:02d}" for i in range(1, 15)
}

actual_notebooks = set(
    P2AUD_C003_UNITS["notebooks"]
)

actual_packages = set(
    P2AUD_C003_UNITS["operational_packages"]
)

missing_notebooks = sorted(
    expected_notebooks - actual_notebooks
)

missing_packages = sorted(
    expected_packages - actual_packages
)

unexpected_notebooks = sorted(
    actual_notebooks - expected_notebooks
)

unexpected_packages = sorted(
    actual_packages - expected_packages
)

master_section_ids = sorted(
    P2AUD_C003_UNITS["master_sections"]
)

P2AUD_C003_STRUCTURE_OK = (
    not missing_notebooks
    and not missing_packages
    and not unexpected_notebooks
    and not unexpected_packages
    and len(master_section_ids) == 2
)

candidate_line_counts = {
    "modular_notebook_items": sum(
        len(unit["candidate_items"])
        for unit in P2AUD_C003_UNITS[
            "notebooks"
        ].values()
    ),
    "operational_package_items": sum(
        len(unit["candidate_items"])
        for unit in P2AUD_C003_UNITS[
            "operational_packages"
        ].values()
    ),
}

P2AUD_C003_REPORT = {
    "schema_version": "P2AUD_ROADMAP_STRUCTURE_V1",
    "cell_id": "P2AUD-C003",
    "action": 3,
    "status": (
        "P2AUD_C003_STRUCTURE_EXTRACTED"
        if P2AUD_C003_STRUCTURE_OK
        else "P2AUD_C003_STRUCTURE_REQUIRES_REVIEW"
    ),
    "reference_versions_verified": True,
    "heading_counts": {
        source_id: len(headings)
        for source_id, headings
        in P2AUD_C003_HEADINGS.items()
    },
    "notebooks_detected": len(
        actual_notebooks
    ),
    "notebook_ids": sorted(
        actual_notebooks
    ),
    "operational_packages_detected": len(
        actual_packages
    ),
    "operational_package_ids": sorted(
        actual_packages
    ),
    "master_sections": master_section_ids,
    "missing_notebooks": missing_notebooks,
    "missing_packages": missing_packages,
    "unexpected_notebooks": unexpected_notebooks,
    "unexpected_packages": unexpected_packages,
    "candidate_line_counts": candidate_line_counts,
    "requirements_evaluated": 0,
    "conflicts_evaluated": 0,
    "scope": "STRUCTURAL_EXTRACTION_ONLY",
    "files_written": 0,
    "new_api_calls": 0,
    "git_operations": 0,
    "next_cell": "P2AUD-C004",
}

assert P2AUD_C003_STRUCTURE_OK, (
    "Roadmap structure requires review"
)

print(
    json.dumps(
        P2AUD_C003_REPORT,
        ensure_ascii=False,
        indent=2,
    )
)
````

## اقدام ۴ — ساخت دفتر نامزدهای الزامات نقشهٔ فاز دوم

### هدف

از روی متن قفل‌شدهٔ سه نقشه، تمام خطوط نامزدِ فعالیت، خروجی، Gate، معیار و شواهد را همراه با مسیر فایل، شناسهٔ بخش، شمارهٔ خط و هش مرجع در حافظه ثبت کنیم.

خروجی‌های موفق سل‌های `P2AUD-C002` و `P2AUD-C003`، بدون بارگذاری مجدد داده و بدون اجرای Pipeline.

### روش

1. استخراج ردیف‌های فهرست، ردیف‌های جدول و خطوط توضیحیِ هر واحد.
2. پوشش NB00–NB23، بسته‌های P2-01 تا P2-14، دو بخش مرجع نقشهٔ کلان و بخش‌های فراگیر خارج از نوت‌بوک‌ها و بسته‌ها.
3. ثبت `candidate_id` پایدار، مسیر و هش منبع، شمارهٔ خط و نوع اولیهٔ هر مورد.
4. نگه‌داشتن موارد مبهم با وضعیت `UNREVIEWED` و `NOT_CLASSIFIED`؛ هیچ تطبیق یا تعارضی هنوز پذیرفته نمی‌شود.
5. ثبت گزارش شمارش‌ها و نمونه‌های قابل‌ردیابی در خروجی سل.

````python
# P2AUD-C004 — Stage traceable roadmap requirement candidates
# Action 4 | READ_ONLY | NO_API | NO_FILE_WRITES | NO_GIT_WRITES

assert P2AUD_C003_REPORT["status"] == "P2AUD_C003_STRUCTURE_EXTRACTED"
assert len(P2AUD_C003_UNITS["notebooks"]) == 24
assert len(P2AUD_C003_UNITS["operational_packages"]) == 14
assert set(P2AUD_C003_UNITS["master_sections"]) == {"PHASE2_SCOPE", "PHASE_HANDOFF"}

# Classification is only a review hint; it is not a compliance judgment.
P2AUD_C004_SECTION_RULES = (
    ("ACCEPTANCE_GATE", ("gate", "پذیرش", "شرط پایان", "شرط ورود")),
    ("DELIVERABLE", ("خروجی", "artifact", "تحویل", "فایل")),
    ("TEST_OR_METRIC", ("معیار", "metric", "test", "آزمون", "تست")),
    ("EVIDENCE", ("شاهد", "evidence", "دانش فنی", "مستندات")),
    ("ACTIVITY", ("سلول", "کارها", "گام", "مراحل", "فعالیت", "variant")),
    ("CONSTRAINT", ("قاعده", "محدودیت", "اصل", "قوانین", "امنیت")),
)


def p2aud_c004_category(section_title):
    normalized = section_title.casefold()
    for category, signals in P2AUD_C004_SECTION_RULES:
        if any(signal in normalized for signal in signals):
            return category
    return "REVIEW_REQUIRED"


def p2aud_c004_global_sections(source_id, excluded_ids):
    lines = P2AUD_ROADMAP_TEXTS[source_id].splitlines()
    headings = P2AUD_C003_HEADINGS[source_id]
    result = {}
    for heading in headings:
        if heading["level"] != 2:
            continue
        title = heading["title"]
        if source_id == "PHASE2_MODULAR" and re.match(
            r"^Notebook\s+\d{2}\s+[—-]", title
        ):
            continue
        if source_id == "PHASE2_OPERATIONAL" and re.search(
            r"\bP2-\d{2}\b", title
        ):
            continue
        key = "GLOBAL-" + str(heading["line"])
        result[key] = p2aud_extract_unit(lines, headings, heading)
    assert not set(result).intersection(excluded_ids)
    return result


P2AUD_C004_GLOBAL_SECTIONS = {
    "PHASE2_MODULAR": p2aud_c004_global_sections(
        "PHASE2_MODULAR", set(P2AUD_C003_UNITS["notebooks"])
    ),
    "PHASE2_OPERATIONAL": p2aud_c004_global_sections(
        "PHASE2_OPERATIONAL", set(P2AUD_C003_UNITS["operational_packages"])
    ),
}

P2AUD_C004_GROUPS = (
    ("PHASE2_MODULAR", "NOTEBOOK", P2AUD_C003_UNITS["notebooks"]),
    ("PHASE2_OPERATIONAL", "PACKAGE", P2AUD_C003_UNITS["operational_packages"]),
    ("MASTER_ROADMAP", "MASTER", P2AUD_C003_UNITS["master_sections"]),
    ("PHASE2_MODULAR", "GLOBAL", P2AUD_C004_GLOBAL_SECTIONS["PHASE2_MODULAR"]),
    ("PHASE2_OPERATIONAL", "GLOBAL", P2AUD_C004_GLOBAL_SECTIONS["PHASE2_OPERATIONAL"]),
)

P2AUD_C004_CANDIDATES = []
P2AUD_C004_SECTION_LEDGER = []
P2AUD_C004_SKIPPED = Counter()
P2AUD_C004_SEEN_IDS = set()

for source_id, unit_type, units in P2AUD_C004_GROUPS:
    source_path = P2AUD_ROADMAP_INDEX[source_id]["relative_path"]
    source_blob_sha = P2AUD_ROADMAP_INDEX[source_id]["git_blob_sha"]
    source_lines = P2AUD_ROADMAP_TEXTS[source_id].splitlines()
    for unit_id, unit in sorted(
        units.items(), key=lambda pair: pair[1]["start_line"]
    ):
        active_section = unit["title"]
        active_category = "REVIEW_REQUIRED"
        in_code_fence = False
        unit_candidate_count = 0
        unit_code_line_count = 0
        unit_uncategorized_count = 0
        for line_no in range(unit["start_line"], unit["end_line"] + 1):
            raw_line = source_lines[line_no - 1]
            content = raw_line.strip()
            if not content:
                continue
            if re.match(r"^\s*(```|~~~)", raw_line):
                in_code_fence = not in_code_fence
                P2AUD_C004_SKIPPED["FENCE_MARKERS"] += 1
                continue
            if in_code_fence:
                unit_code_line_count += 1
                P2AUD_C004_SKIPPED["CODE_EXAMPLE_LINES"] += 1
                continue
            heading_match = re.match(r"^#{1,6}\s+(.+?)\s*$", raw_line)
            if heading_match:
                active_section = heading_match.group(1)
                active_category = p2aud_c004_category(active_section)
                continue
            if re.fullmatch(r"[-_*]{3,}", content):
                continue
            if content.startswith(("<div", "</div", "<table", "</table")):
                P2AUD_C004_SKIPPED["HTML_LAYOUT"] += 1
                continue
            if re.match(r"^\|\s*:?-{2,}", content):
                P2AUD_C004_SKIPPED["TABLE_SEPARATOR"] += 1
                continue

            if re.match(r"^\s*(?:[-*+]\s+|\d+[.)]\s+)", raw_line):
                record_kind = "LIST_ITEM"
            elif content.startswith("|") and content.endswith("|"):
                record_kind = "TABLE_ROW"
            else:
                record_kind = "PARAGRAPH_LINE"
                if active_category == "REVIEW_REQUIRED":
                    unit_uncategorized_count += 1

            # Keep possible explanatory lines visible; never silently discard them.
            candidate_key = (
                f"{source_id}|{unit_type}|{unit_id}|{line_no}|{raw_line}"
            )
            candidate_id = (
                "P2AUD-RC-" + hashlib.sha256(
                    candidate_key.encode("utf-8")
                ).hexdigest()[:16]
            )
            if candidate_id in P2AUD_C004_SEEN_IDS:
                raise AssertionError(f"Duplicate candidate ID: {candidate_id}")
            P2AUD_C004_SEEN_IDS.add(candidate_id)
            P2AUD_C004_CANDIDATES.append({
                "candidate_id": candidate_id,
                "source_id": source_id,
                "source_path": source_path,
                "source_blob_sha": source_blob_sha,
                "unit_type": unit_type,
                "unit_id": unit_id,
                "unit_title": unit["title"],
                "section_title": active_section,
                "category_hint": active_category,
                "record_kind": record_kind,
                "source_line": line_no,
                "text": content,
                "review_status": "UNREVIEWED",
                "requirement_status": "NOT_CLASSIFIED",
                "conflict_status": "NOT_EVALUATED",
            })
            unit_candidate_count += 1
        if in_code_fence:
            raise ValueError(
                f"Unclosed Markdown fence in {source_id}/{unit_id}"
            )
        P2AUD_C004_SECTION_LEDGER.append({
            "source_id": source_id,
            "unit_type": unit_type,
            "unit_id": unit_id,
            "start_line": unit["start_line"],
            "end_line": unit["end_line"],
            "section_sha256": unit["section_sha256"],
            "candidate_count": unit_candidate_count,
            "code_example_lines_excluded": unit_code_line_count,
            "uncategorized_paragraph_lines": unit_uncategorized_count,
        })

# Verify that every staged item traces to the immutable source inventory.
for record in P2AUD_C004_CANDIDATES:
    source_line = P2AUD_ROADMAP_TEXTS[
        record["source_id"]
    ].splitlines()[record["source_line"] - 1].strip()
    assert source_line == record["text"], record["candidate_id"]

P2AUD_C004_BY_KIND = Counter(
    item["record_kind"] for item in P2AUD_C004_CANDIDATES
)
P2AUD_C004_BY_CATEGORY = Counter(
    item["category_hint"] for item in P2AUD_C004_CANDIDATES
)
P2AUD_C004_BY_SOURCE = Counter(
    item["source_id"] for item in P2AUD_C004_CANDIDATES
)
P2AUD_C004_BY_UNIT_TYPE = Counter(
    item["unit_type"] for item in P2AUD_C004_CANDIDATES
)

assert len(P2AUD_C004_CANDIDATES) == len(P2AUD_C004_SEEN_IDS)
assert sum(P2AUD_C004_BY_SOURCE.values()) == len(P2AUD_C004_CANDIDATES)
assert len(P2AUD_C004_SECTION_LEDGER) >= 40
assert all(
    item["review_status"] == "UNREVIEWED"
    for item in P2AUD_C004_CANDIDATES
)

P2AUD_C004_SAMPLE = []
for sample_type in ("NOTEBOOK", "PACKAGE", "MASTER", "GLOBAL"):
    example = next(
        (x for x in P2AUD_C004_CANDIDATES
         if x["unit_type"] == sample_type),
        None,
    )
    if example:
        P2AUD_C004_SAMPLE.append({
            "candidate_id": example["candidate_id"],
            "source_id": example["source_id"],
            "unit_id": example["unit_id"],
            "source_line": example["source_line"],
            "category_hint": example["category_hint"],
            "text_preview": example["text"][:135],
        })

P2AUD_C004_REPORT = {
    "schema_version": "P2AUD_ROADMAP_CANDIDATE_INVENTORY_V1",
    "cell_id": "P2AUD-C004",
    "action": 4,
    "status": "P2AUD_C004_CANDIDATES_STAGED_NOT_ADJUDICATED",
    "source_versions_match_reference": True,
    "units_scanned": len(P2AUD_C004_SECTION_LEDGER),
    "global_sections_scanned": sum(
        len(parts) for parts in P2AUD_C004_GLOBAL_SECTIONS.values()
    ),
    "candidate_count": len(P2AUD_C004_CANDIDATES),
    "counts_by_source": dict(sorted(P2AUD_C004_BY_SOURCE.items())),
    "counts_by_unit_type": dict(sorted(P2AUD_C004_BY_UNIT_TYPE.items())),
    "counts_by_record_kind": dict(sorted(P2AUD_C004_BY_KIND.items())),
    "counts_by_category_hint": dict(sorted(P2AUD_C004_BY_CATEGORY.items())),
    "skipped_non_candidate_lines": dict(sorted(P2AUD_C004_SKIPPED.items())),
    "sample_candidates": P2AUD_C004_SAMPLE,
    "requirements_evaluated": 0,
    "conflicts_evaluated": 0,
    "files_written": 0,
    "new_api_calls": 0,
    "git_operations": 0,
    "next_cell": "P2AUD-C005",
}

print(json.dumps(P2AUD_C004_REPORT, ensure_ascii=False, indent=2))
````

````python
# P2AUD-C005 — Index Roadmap Gates, Deliverables and Evidence
# Action 5 | READ_ONLY | NO_API | NO_FILE_WRITES | NO_GIT_WRITES

assert P2AUD_C004_REPORT["status"] == (
    "P2AUD_C004_CANDIDATES_STAGED_NOT_ADJUDICATED"
)
assert len(P2AUD_C004_CANDIDATES) == 654

P2AUD_C005_ROLE_RULES = {
    "ACCEPTANCE_GATE": (
        r"\bgate\b",
        r"پذیرش",
        r"تعریف\s+done",
        r"شرط\s+پایان",
        r"شرط\s+ورود",
    ),
    "DELIVERABLE": (
        r"خروجی",
        r"artifact",
        r"deliverable",
        r"تحویل",
    ),
    "TEST_OR_METRIC": (
        r"معیار",
        r"metric",
        r"benchmark",
        r"آزمون",
        r"تست",
        r"ارزیابی",
        r"regression",
    ),
    "EVIDENCE": (
        r"شاهد",
        r"evidence",
        r"مستندات",
        r"دانش\s+فنی",
    ),
    "ACTIVITY": (
        r"سلول",
        r"فعالیت",
        r"کارها",
        r"گام",
        r"مراحل",
        r"ترتیب\s+اجرا",
    ),
    "CONSTRAINT": (
        r"قاعده",
        r"محدودیت",
        r"اصول",
        r"قوانین",
    ),
}


def p2aud_c005_heading_roles(title):
    normalized = unicodedata.normalize(
        "NFKC", title
    ).casefold()

    return [
        role
        for role, expressions in P2AUD_C005_ROLE_RULES.items()
        if any(
            re.search(expression, normalized, re.IGNORECASE)
            for expression in expressions
        )
    ]


P2AUD_C005_CANDIDATES_BY_UNIT = defaultdict(list)

for candidate in P2AUD_C004_CANDIDATES:
    key = (
        candidate["source_id"],
        candidate["unit_type"],
        candidate["unit_id"],
    )
    P2AUD_C005_CANDIDATES_BY_UNIT[key].append(candidate)

P2AUD_C005_ANCHORS = []
P2AUD_C005_UNIT_LEDGER = []
P2AUD_C005_REVIEW_FLAGS = []

for source_id, unit_type, units in P2AUD_C004_GROUPS:
    headings = P2AUD_C003_HEADINGS[source_id]
    source_info = P2AUD_ROADMAP_INDEX[source_id]

    for unit_id, unit in sorted(
        units.items(),
        key=lambda item: item[1]["start_line"],
    ):
        start = unit["start_line"]
        end = unit["end_line"]

        unit_headings = [
            heading
            for heading in headings
            if start <= heading["line"] <= end
        ]

        unit_candidates = P2AUD_C005_CANDIDATES_BY_UNIT.get(
            (source_id, unit_type, unit_id),
            [],
        )

        unit_anchors = []
        role_counter = Counter()

        for heading in unit_headings:
            roles = p2aud_c005_heading_roles(
                heading["title"]
            )

            if not roles:
                continue

            heading_start = heading["line"]
            heading_end = end

            for next_heading in unit_headings:
                if (
                    next_heading["line"] > heading_start
                    and next_heading["level"] <= heading["level"]
                ):
                    heading_end = next_heading["line"] - 1
                    break

            related_candidates = [
                item
                for item in unit_candidates
                if (
                    heading_start
                    <= item["source_line"]
                    <= heading_end
                )
            ]

            identity = (
                f"{source_id}|{unit_type}|{unit_id}|"
                f"{heading_start}|{heading['title']}|"
                f"{source_info['git_blob_sha']}"
            )

            anchor_id = (
                "P2AUD-AN-"
                + hashlib.sha256(
                    identity.encode("utf-8")
                ).hexdigest()[:16]
            )

            anchor = {
                "anchor_id": anchor_id,
                "source_id": source_id,
                "source_path": source_info["relative_path"],
                "source_blob_sha": source_info["git_blob_sha"],
                "unit_type": unit_type,
                "unit_id": unit_id,
                "heading": heading["title"],
                "heading_level": heading["level"],
                "start_line": heading_start,
                "end_line": heading_end,
                "role_hints": roles,
                "related_candidate_count": len(
                    related_candidates
                ),
                "related_candidate_ids": [
                    item["candidate_id"]
                    for item in related_candidates
                ],
                "examples": [
                    {
                        "line": item["source_line"],
                        "text": item["text"][:160],
                    }
                    for item in related_candidates[:2]
                ],
                "review_status": "UNREVIEWED",
                "gate_result": "NOT_EVALUATED",
            }

            P2AUD_C005_ANCHORS.append(anchor)
            unit_anchors.append(anchor)

            for role in roles:
                role_counter[role] += 1

        if (
            unit_type in ("NOTEBOOK", "PACKAGE")
            and role_counter["ACCEPTANCE_GATE"] == 0
        ):
            P2AUD_C005_REVIEW_FLAGS.append({
                "source_id": source_id,
                "unit_type": unit_type,
                "unit_id": unit_id,
                "flag": "NO_EXPLICIT_ACCEPTANCE_HEADING",
                "interpretation": (
                    "REVIEW_REQUIRED_NOT_A_GATE_FAILURE"
                ),
            })

        P2AUD_C005_UNIT_LEDGER.append({
            "source_id": source_id,
            "unit_type": unit_type,
            "unit_id": unit_id,
            "unit_title": unit["title"],
            "start_line": start,
            "end_line": end,
            "candidate_count": len(unit_candidates),
            "anchor_count": len(unit_anchors),
            "anchor_roles": dict(sorted(
                role_counter.items()
            )),
            "anchor_ids": [
                anchor["anchor_id"]
                for anchor in unit_anchors
            ],
            "status": "STRUCTURE_INDEXED_NOT_ADJUDICATED",
        })

anchor_ids = [
    item["anchor_id"]
    for item in P2AUD_C005_ANCHORS
]

assert len(anchor_ids) == len(set(anchor_ids)), (
    "Duplicate anchor identifiers detected"
)

assert len(P2AUD_C005_UNIT_LEDGER) == len(
    P2AUD_C004_SECTION_LEDGER
), "Unit coverage differs from C004"

assert sum(
    item["candidate_count"]
    for item in P2AUD_C005_UNIT_LEDGER
) == len(P2AUD_C004_CANDIDATES), (
    "Candidate coverage mismatch"
)

P2AUD_C005_ROLE_COUNTS = Counter()

for anchor in P2AUD_C005_ANCHORS:
    for role in anchor["role_hints"]:
        P2AUD_C005_ROLE_COUNTS[role] += 1

P2AUD_C005_SOURCE_COUNTS = Counter(
    anchor["source_id"]
    for anchor in P2AUD_C005_ANCHORS
)

P2AUD_C005_NO_GATE_BY_TYPE = Counter(
    flag["unit_type"]
    for flag in P2AUD_C005_REVIEW_FLAGS
)

P2AUD_C005_GATE_SAMPLES = [
    {
        "source_id": anchor["source_id"],
        "unit_id": anchor["unit_id"],
        "heading": anchor["heading"],
        "start_line": anchor["start_line"],
        "end_line": anchor["end_line"],
        "candidate_count": anchor["related_candidate_count"],
        "example": (
            anchor["examples"][0]
            if anchor["examples"]
            else None
        ),
    }
    for anchor in P2AUD_C005_ANCHORS
    if "ACCEPTANCE_GATE" in anchor["role_hints"]
][:8]

P2AUD_C005_REPORT = {
    "schema_version": "P2AUD_ROADMAP_CONTROL_ANCHORS_V1",
    "cell_id": "P2AUD-C005",
    "action": 5,
    "status": (
        "P2AUD_C005_CONTROL_ANCHORS_INDEXED_NOT_ADJUDICATED"
    ),
    "roadmaps_loaded": len(P2AUD_ROADMAP_TEXTS),
    "units_indexed": len(P2AUD_C005_UNIT_LEDGER),
    "candidates_preserved": len(P2AUD_C004_CANDIDATES),
    "anchors_detected": len(P2AUD_C005_ANCHORS),
    "anchor_counts_by_source": dict(sorted(
        P2AUD_C005_SOURCE_COUNTS.items()
    )),
    "anchor_counts_by_role": dict(sorted(
        P2AUD_C005_ROLE_COUNTS.items()
    )),
    "units_without_explicit_gate_heading": len(
        P2AUD_C005_REVIEW_FLAGS
    ),
    "no_gate_heading_by_unit_type": dict(sorted(
        P2AUD_C005_NO_GATE_BY_TYPE.items()
    )),
    "review_flags": P2AUD_C005_REVIEW_FLAGS,
    "sample_acceptance_anchors": P2AUD_C005_GATE_SAMPLES,
    "requirements_evaluated": 0,
    "conflicts_evaluated": 0,
    "formal_gates_passed": 0,
    "files_written": 0,
    "new_api_calls": 0,
    "git_operations": 0,
    "next_cell": "P2AUD-C006",
}

assert P2AUD_C005_REPORT["units_indexed"] == 64
assert P2AUD_C005_REPORT["candidates_preserved"] == 654
assert P2AUD_C005_REPORT["formal_gates_passed"] == 0

print(
    json.dumps(
        P2AUD_C005_REPORT,
        ensure_ascii=False,
        indent=2,
    )
)
````

بررسی رابطهٔ واقعی بین نقشهٔ عملیاتی و نقشهٔ ماژولار فاز دوم، با حفظ تمام تفاوت‌ها و شواهد.

````python
# P2AUD-C012 — Inspect saved notebook evidence and artifact locations
# Action 12 | READ_ONLY | NO_API | NO_FILE_WRITES | NO_GIT_WRITES

assert P2AUD_C011_REPORT["status"] in {
    "P2AUD_C011_INVENTORY_COMPLETE",
    "P2AUD_C011_INVENTORY_COMPLETE_WITH_REVIEW_FLAGS",
}
assert len(P2AUD_C011_NOTEBOOK_REGISTRY) == 24
assert len(P2AUD_C010_CASE_DOSSIERS) == 5

P2AUD_C012_ROOT = Path(P2AUD_REPO_ROOT).resolve()
P2AUD_C012_MAX_NOTEBOOK_BYTES = 64 * 1024 * 1024
P2AUD_C012_MAX_ARTIFACT_HASH_BYTES = 4 * 1024 * 1024

P2AUD_C012_PATTERNS = {
    "P2AUD-RV-001": (
        r"citation.{0,40}coverage|"
        r"claim.{0,50}citation|"
        r"citation.{0,50}claim|"
        r"citation.{0,30}correctness"
    ),
    "P2AUD-RV-002": (
        r"schema|contract|lineage|"
        r"stable[_-]?id|migration"
    ),
    "P2AUD-RV-003": (
        r"golden|chunk|locator|"
        r"annotation|gold[_-]?documents"
    ),
    "P2AUD-RV-004": (
        r"taxonomy|competency|topic|ontology"
    ),
    "P2AUD-RV-005": (
        r"handoff|rag_request|rag_response|"
        r"readiness|backward[_ -]?compatibility|"
        r"release[_ -]?manifest"
    ),
}

P2AUD_C012_COMPILED = {
    review_id: re.compile(pattern, re.IGNORECASE)
    for review_id, pattern in P2AUD_C012_PATTERNS.items()
}

P2AUD_C012_FOCUS_BY_NOTEBOOK = defaultdict(list)

for review_id, notebook_ids in P2AUD_C011_FOCUS.items():
    assert review_id in P2AUD_C012_COMPILED

    for notebook_id in notebook_ids:
        assert notebook_id in P2AUD_C011_NOTEBOOK_REGISTRY
        P2AUD_C012_FOCUS_BY_NOTEBOOK[
            notebook_id
        ].append(review_id)


def p2aud_c012_source(cell):
    value = cell.get("source", "")

    if isinstance(value, str):
        return value

    if isinstance(value, list):
        return "".join(
            part for part in value
            if isinstance(part, str)
        )

    return ""


P2AUD_C012_NOTEBOOK_EVIDENCE = {}
P2AUD_C012_STATIC_CODE_EVIDENCE = []

for notebook_id, registered in sorted(
    P2AUD_C011_NOTEBOOK_REGISTRY.items()
):
    relative_path = registered["selected_path"]

    if (
        not relative_path
        or registered["selected_read_status"] != "READ_SUCCESS"
    ):
        raise ValueError(
            f"Notebook is not readable: {notebook_id}"
        )

    path = (P2AUD_C012_ROOT / relative_path).resolve()

    assert path.is_relative_to(P2AUD_C012_ROOT)
    assert path.is_file()

    if path.stat().st_size > P2AUD_C012_MAX_NOTEBOOK_BYTES:
        raise ValueError(
            f"Notebook exceeds read limit: {notebook_id}"
        )

    raw = path.read_bytes()
    current_sha256 = hashlib.sha256(raw).hexdigest()

    assert current_sha256 == registered[
        "selected_file_sha256"
    ], f"Notebook changed after C011: {notebook_id}"

    notebook = json.loads(raw.decode("utf-8-sig"))
    cells = notebook.get("cells")

    assert isinstance(cells, list)

    unrun_cell_numbers = []
    error_records = []
    code_count = 0

    code_matches = {
        review_id: {
            "matched_code_cells": 0,
            "keyword_line_matches": 0,
            "examples": [],
        }
        for review_id in P2AUD_C012_FOCUS_BY_NOTEBOOK[
            notebook_id
        ]
    }

    for cell_number, cell in enumerate(cells, start=1):
        if cell.get("cell_type") != "code":
            continue

        code_count += 1
        source = p2aud_c012_source(cell)

        if (
            source.strip()
            and cell.get("execution_count") is None
        ):
            unrun_cell_numbers.append(cell_number)

        outputs = cell.get("outputs", [])

        if not isinstance(outputs, list):
            outputs = []

        for output in outputs:
            if (
                isinstance(output, dict)
                and output.get("output_type") == "error"
            ):
                error_records.append({
                    "cell_number": cell_number,
                    "error_type": str(
                        output.get("ename", "UNKNOWN")
                    )[:100],
                    "execution_count": cell.get(
                        "execution_count"
                    ),
                })

        if not source.strip():
            continue

        source_lines = source.splitlines()

        for review_id, evidence in code_matches.items():
            regex = P2AUD_C012_COMPILED[review_id]
            cell_has_match = False

            for line_number, line in enumerate(
                source_lines, start=1
            ):
                match = regex.search(line)

                if not match:
                    continue

                cell_has_match = True
                evidence["keyword_line_matches"] += 1

                if len(evidence["examples"]) < 3:
                    evidence["examples"].append({
                        "cell_number": cell_number,
                        "line_in_cell": line_number,
                        "matched_term": match.group(0)[:60],
                    })

            if cell_has_match:
                evidence["matched_code_cells"] += 1

    assert code_count == registered["saved_code_cells"], (
        f"Code-cell count changed: {notebook_id}"
    )

    assert len(unrun_cell_numbers) == registered[
        "saved_unrun_code_cells"
    ], f"Unrun count changed: {notebook_id}"

    assert len(error_records) == registered[
        "saved_error_outputs"
    ], f"Error count changed: {notebook_id}"

    P2AUD_C012_NOTEBOOK_EVIDENCE[notebook_id] = {
        "relative_path": relative_path,
        "sha256": current_sha256,
        "code_cells": code_count,
        "saved_unrun_count": len(unrun_cell_numbers),
        "unrun_cell_examples": unrun_cell_numbers[:12],
        "saved_error_count": len(error_records),
        "error_records": error_records[:12],
        "execution_proof": "NOT_VERIFIED_BY_STATIC_INSPECTION",
        "formal_acceptance": "NOT_EVALUATED",
    }

    for review_id, evidence in code_matches.items():
        P2AUD_C012_STATIC_CODE_EVIDENCE.append({
            "review_id": review_id,
            "notebook_id": notebook_id,
            "notebook_path": relative_path,
            "notebook_sha256": current_sha256,
            **evidence,
            "evidence_status": (
                "STATIC_CODE_TOKEN_HITS_NOT_EXECUTION_PROOF"
            ),
        })


# Check exact roadmap-defined artifact locations.
P2AUD_C012_ARTIFACT_TARGETS = {
    "RAG_REQUEST_SCHEMA": (
        "schemas/phase2/rag_request.schema.json", "FILE"
    ),
    "RAG_RESPONSE_SCHEMA": (
        "schemas/phase2/rag_response.schema.json", "FILE"
    ),
    "PHASE3_READINESS_REPORT": (
        "reports/phase2/release/phase3_readiness.md", "FILE"
    ),
    "PHASE3_REQUEST_TEST": (
        "tests/phase2/contracts/test_phase3_request.py", "FILE"
    ),
    "PHASE3_RESPONSE_TEST": (
        "tests/phase2/contracts/test_phase3_response.py", "FILE"
    ),
    "VERSION_TRACE_TEST": (
        "tests/phase2/contracts/test_version_trace.py", "FILE"
    ),
    "DATA_MANIFESTS": (
        "data/phase2/manifests", "DIRECTORY"
    ),
    "DATA_PROCESSED": (
        "data/phase2/processed", "DIRECTORY"
    ),
    "CHUNK_ARTIFACTS": (
        "data/phase2/chunks", "DIRECTORY"
    ),
    "INDEX_ARTIFACTS": (
        "data/phase2/index", "DIRECTORY"
    ),
    "SCHEMAS": (
        "schemas/phase2", "DIRECTORY"
    ),
    "CONFIG": (
        "config/phase2", "DIRECTORY"
    ),
    "REPORTS": (
        "reports/phase2", "DIRECTORY"
    ),
    "CHECKPOINTS": (
        "artifacts/phase2/checkpoints", "DIRECTORY"
    ),
    "RELEASES": (
        "artifacts/phase2/releases", "DIRECTORY"
    ),
    "CONTRACT_TESTS": (
        "tests/phase2/contracts", "DIRECTORY"
    ),
}

P2AUD_C012_ARTIFACT_INVENTORY = {}

for artifact_id, (relative_path, expected_type) in (
    P2AUD_C012_ARTIFACT_TARGETS.items()
):
    path = P2AUD_C012_ROOT / relative_path

    record = {
        "artifact_id": artifact_id,
        "relative_path": relative_path,
        "expected_type": expected_type,
        "status": "NOT_FOUND_AT_PLANNED_PATH",
        "size_bytes": None,
        "sha256": None,
        "direct_child_examples": [],
        "direct_children_sampled": 0,
        "review_status": "NOT_EVALUATED",
    }

    if path.is_symlink():
        record["status"] = "SYMLINK_REQUIRES_REVIEW"

    elif expected_type == "FILE" and path.is_file():
        try:
            size = path.stat().st_size
            record["size_bytes"] = size

            if size <= P2AUD_C012_MAX_ARTIFACT_HASH_BYTES:
                record["sha256"] = hashlib.sha256(
                    path.read_bytes()
                ).hexdigest()
                record["status"] = "FILE_FOUND_HASHED"
            else:
                record["status"] = "FILE_FOUND_HASH_SKIPPED_SIZE"

        except OSError as exc:
            record["status"] = "FILE_READ_ERROR"
            record["error_type"] = type(exc).__name__

    elif expected_type == "DIRECTORY" and path.is_dir():
        record["status"] = "DIRECTORY_FOUND"

        try:
            with os.scandir(path) as entries:
                for index, entry in enumerate(entries):
                    if index >= 40:
                        record["status"] = (
                            "DIRECTORY_FOUND_LISTING_TRUNCATED"
                        )
                        break

                    record["direct_children_sampled"] += 1

                    if len(record["direct_child_examples"]) < 8:
                        record["direct_child_examples"].append({
                            "name": entry.name[:120],
                            "kind": (
                                "DIRECTORY"
                                if entry.is_dir(follow_symlinks=False)
                                else "FILE_OR_OTHER"
                            ),
                        })

        except OSError as exc:
            record["status"] = "DIRECTORY_LIST_ERROR"
            record["error_type"] = type(exc).__name__

    elif path.exists():
        record["status"] = "WRONG_OBJECT_TYPE"

    P2AUD_C012_ARTIFACT_INVENTORY[
        artifact_id
    ] = record


# Release IDs are not guessed. Scan only immediate release folders.
P2AUD_C012_RELEASE_INVENTORY = {
    "release_root": "artifacts/phase2/releases",
    "status": "NOT_FOUND",
    "release_directories_scanned": 0,
    "scan_truncated": False,
    "handoff_contracts_found": [],
    "release_directory_examples": [],
}

release_root = (
    P2AUD_C012_ROOT / "artifacts/phase2/releases"
)

if release_root.is_dir() and not release_root.is_symlink():
    P2AUD_C012_RELEASE_INVENTORY["status"] = "SCANNED"

    try:
        with os.scandir(release_root) as entries:
            for index, entry in enumerate(entries):
                if index >= 100:
                    P2AUD_C012_RELEASE_INVENTORY[
                        "scan_truncated"
                    ] = True
                    break

                if not entry.is_dir(follow_symlinks=False):
                    continue

                P2AUD_C012_RELEASE_INVENTORY[
                    "release_directories_scanned"
                ] += 1

                if len(
                    P2AUD_C012_RELEASE_INVENTORY[
                        "release_directory_examples"
                    ]
                ) < 12:
                    P2AUD_C012_RELEASE_INVENTORY[
                        "release_directory_examples"
                    ].append(entry.name[:120])

                contract_path = (
                    Path(entry.path)
                    / "phase3_handoff_contract.json"
                )

                if (
                    not contract_path.is_symlink()
                    and contract_path.is_file()
                ):
                    P2AUD_C012_RELEASE_INVENTORY[
                        "handoff_contracts_found"
                    ].append(
                        contract_path.relative_to(
                            P2AUD_C012_ROOT
                        ).as_posix()
                    )

    except OSError as exc:
        P2AUD_C012_RELEASE_INVENTORY[
            "status"
        ] = "SCAN_ERROR"
        P2AUD_C012_RELEASE_INVENTORY[
            "error_type"
        ] = type(exc).__name__


P2AUD_C012_NOTEBOOK_FLAGS = {
    notebook_id: {
        "saved_unrun_count": item["saved_unrun_count"],
        "unrun_cell_examples": item["unrun_cell_examples"],
        "saved_error_count": item["saved_error_count"],
        "error_records": item["error_records"],
    }
    for notebook_id, item in (
        P2AUD_C012_NOTEBOOK_EVIDENCE.items()
    )
    if (
        item["saved_unrun_count"] > 0
        or item["saved_error_count"] > 0
    )
}

P2AUD_C012_ARTIFACT_STATUS_COUNTS = Counter(
    record["status"]
    for record in P2AUD_C012_ARTIFACT_INVENTORY.values()
)

P2AUD_C012_CODE_EVIDENCE_SUMMARY = []

for review_id in sorted(P2AUD_C012_COMPILED):
    matches = [
        item
        for item in P2AUD_C012_STATIC_CODE_EVIDENCE
        if item["review_id"] == review_id
    ]

    P2AUD_C012_CODE_EVIDENCE_SUMMARY.append({
        "review_id": review_id,
        "notebooks_examined": len(matches),
        "notebooks_with_token_hits": [
            item["notebook_id"]
            for item in matches
            if item["keyword_line_matches"] > 0
        ],
        "total_keyword_line_matches": sum(
            item["keyword_line_matches"]
            for item in matches
        ),
        "sample_locations": [
            {
                "notebook_id": item["notebook_id"],
                **example,
            }
            for item in matches
            for example in item["examples"][:1]
        ][:8],
        "implementation_verified": False,
    })


P2AUD_C012_REPORT = {
    "schema_version": (
        "P2AUD_STATIC_EXECUTION_AND_ARTIFACT_EVIDENCE_V1"
    ),
    "cell_id": "P2AUD-C012",
    "action": 12,
    "status": "P2AUD_C012_EVIDENCE_STAGED_REVIEW_REQUIRED",
    "planned_notebooks_reverified": len(
        P2AUD_C012_NOTEBOOK_EVIDENCE
    ),
    "notebooks_with_saved_execution_flags": len(
        P2AUD_C012_NOTEBOOK_FLAGS
    ),
    "saved_execution_flags": P2AUD_C012_NOTEBOOK_FLAGS,
    "code_evidence_summary": (
        P2AUD_C012_CODE_EVIDENCE_SUMMARY
    ),
    "planned_artifact_paths_checked": len(
        P2AUD_C012_ARTIFACT_INVENTORY
    ),
    "artifact_status_counts": dict(
        sorted(P2AUD_C012_ARTIFACT_STATUS_COUNTS.items())
    ),
    "artifact_file_status": {
        artifact_id: {
            "status": record["status"],
            "relative_path": record["relative_path"],
            "sha256": record["sha256"],
        }
        for artifact_id, record in (
            P2AUD_C012_ARTIFACT_INVENTORY.items()
        )
        if record["expected_type"] == "FILE"
    },
    "release_inventory": P2AUD_C012_RELEASE_INVENTORY,
    "additional_notebooks_from_c011": len(
        P2AUD_C011_VARIANT_REGISTRY
    ),
    "additional_notebooks_formally_evaluated": 0,
    "requirements_implementation_accepted": 0,
    "final_decisions_approved": 0,
    "formal_gates_passed": 0,
    "files_written": 0,
    "new_api_calls": 0,
    "git_operations": 0,
    "next_cell": "P2AUD-C013",
}

assert P2AUD_C012_REPORT[
    "planned_notebooks_reverified"
] == 24

assert P2AUD_C012_REPORT[
    "planned_artifact_paths_checked"
] == 16

assert P2AUD_C012_REPORT[
    "final_decisions_approved"
] == 0

print(
    json.dumps(
        P2AUD_C012_REPORT,
        ensure_ascii=False,
        indent=2,
    )
)
````

````python
# P2AUD-C013 — Reconcile missing artifacts and notebook variants
# Action 13 | READ_ONLY | DRIFT_AWARE | NO_API | NO_GIT_WRITES

assert P2AUD_C012_REPORT["status"] == (
    "P2AUD_C012_EVIDENCE_STAGED_REVIEW_REQUIRED"
)
assert len(P2AUD_C011_VARIANT_REGISTRY) == 26

P2AUD_C013_ROOT = Path(P2AUD_REPO_ROOT).resolve()
P2AUD_C013_MAX_BYTES = 64 * 1024 * 1024

P2AUD_C013_MISSING_TARGETS = {
    artifact_id: record
    for artifact_id, record in P2AUD_C012_ARTIFACT_INVENTORY.items()
    if record["status"] == "NOT_FOUND_AT_PLANNED_PATH"
}

assert len(P2AUD_C013_MISSING_TARGETS) == 6

P2AUD_C013_SEARCH_ROOTS = (
    "reports",
    "artifacts",
    "tests",
    "schemas",
    "src/phase2",
    "notebooks/phase2",
    "data/phase2",
    "config/phase2",
    "docs/phase2",
)

P2AUD_C013_SKIP_DIRECTORIES = {
    ".git", ".venv", "venv", "node_modules",
    "__pycache__", ".ipynb_checkpoints",
    ".pytest_cache", ".mypy_cache",
    "raw", "interim", "processed",
    "chunks", "embeddings", "indexes",
    "qdrant_storage",
}

P2AUD_C013_SCANNED_FILES = []
P2AUD_C013_SCANNED_DIRECTORIES = []
P2AUD_C013_SCAN_ROOT_STATUS = []
P2AUD_C013_SCAN_TRUNCATED = False

MAX_FILES = 30000
MAX_DIRS = 10000

for relative_root in P2AUD_C013_SEARCH_ROOTS:
    search_root = P2AUD_C013_ROOT / relative_root

    if search_root.is_symlink():
        P2AUD_C013_SCAN_ROOT_STATUS.append({
            "root": relative_root,
            "status": "SKIPPED_SYMLINK",
        })
        continue

    if not search_root.is_dir():
        P2AUD_C013_SCAN_ROOT_STATUS.append({
            "root": relative_root,
            "status": "NOT_FOUND",
        })
        continue

    P2AUD_C013_SCAN_ROOT_STATUS.append({
        "root": relative_root,
        "status": "SCANNED",
    })

    for current_dir, subdirs, filenames in os.walk(
        search_root,
        topdown=True,
        followlinks=False,
    ):
        current = Path(current_dir)

        subdirs[:] = sorted(
            name
            for name in subdirs
            if name not in P2AUD_C013_SKIP_DIRECTORIES
            and not (current / name).is_symlink()
        )

        relative_dir = current.relative_to(
            P2AUD_C013_ROOT
        ).as_posix()

        P2AUD_C013_SCANNED_DIRECTORIES.append(
            relative_dir
        )

        for filename in sorted(filenames):
            file_path = current / filename

            if file_path.is_symlink() or not file_path.is_file():
                continue

            P2AUD_C013_SCANNED_FILES.append(
                file_path.relative_to(
                    P2AUD_C013_ROOT
                ).as_posix()
            )

            if len(P2AUD_C013_SCANNED_FILES) >= MAX_FILES:
                P2AUD_C013_SCAN_TRUNCATED = True
                break

        if (
            P2AUD_C013_SCAN_TRUNCATED
            or len(P2AUD_C013_SCANNED_DIRECTORIES) >= MAX_DIRS
        ):
            P2AUD_C013_SCAN_TRUNCATED = True
            break

    if P2AUD_C013_SCAN_TRUNCATED:
        break


# Locate missing planned artifacts by exact filename or directory name.

P2AUD_C013_ALTERNATIVE_LOCATIONS = {}

for artifact_id, missing_record in sorted(
    P2AUD_C013_MISSING_TARGETS.items()
):
    expected_path = missing_record["relative_path"]
    expected_name = Path(expected_path).name.casefold()
    expected_type = missing_record["expected_type"]

    candidates = (
        P2AUD_C013_SCANNED_FILES
        if expected_type == "FILE"
        else P2AUD_C013_SCANNED_DIRECTORIES
    )

    matches = [
        relative_path
        for relative_path in candidates
        if (
            Path(relative_path).name.casefold() == expected_name
            and relative_path != expected_path
        )
    ]

    P2AUD_C013_ALTERNATIVE_LOCATIONS[artifact_id] = {
        "expected_path": expected_path,
        "expected_type": expected_type,
        "exact_name_alternatives_count": len(matches),
        "exact_name_alternatives": matches[:20],
        "search_status": (
            "ALTERNATIVE_NAME_MATCH_REQUIRES_VALIDATION"
            if matches
            else (
                "SCAN_INCOMPLETE"
                if P2AUD_C013_SCAN_TRUNCATED
                else "NOT_FOUND_IN_BOUNDED_SCAN"
            )
        ),
        "content_validated": False,
        "formal_acceptance": "NOT_EVALUATED",
    }


P2AUD_C013_RELEASE_PATTERN = re.compile(
    r"release|readiness|handoff|phase.?3|version.?trace",
    re.IGNORECASE,
)

P2AUD_C013_RELEASE_FILE_CANDIDATES = [
    relative_path
    for relative_path in P2AUD_C013_SCANNED_FILES
    if P2AUD_C013_RELEASE_PATTERN.search(
        Path(relative_path).name
    )
    and Path(relative_path).suffix.lower() in {
        ".json", ".md", ".yaml", ".yml", ".py", ".txt",
    }
]

P2AUD_C013_RELEASE_DIR_CANDIDATES = [
    relative_path
    for relative_path in P2AUD_C013_SCANNED_DIRECTORIES
    if re.search(
        r"release|handoff|readiness",
        Path(relative_path).name,
        re.IGNORECASE,
    )
]


# Re-fingerprint variant notebooks without modifying them.
# A changed hash is recorded, not treated as an automatic failure.

def p2aud_c013_read_snapshot(relative_path):
    path = P2AUD_C013_ROOT / relative_path

    result = {
        "read_status": "NOT_READ",
        "sha256": None,
        "size_bytes": None,
        "raw": None,
    }

    if path.is_symlink():
        result["read_status"] = "SYMLINK_REQUIRES_REVIEW"
        return result

    if not path.is_file():
        result["read_status"] = "FILE_NOT_FOUND"
        return result

    if not path.resolve().is_relative_to(P2AUD_C013_ROOT):
        result["read_status"] = "OUTSIDE_REPOSITORY"
        return result

    try:
        size = path.stat().st_size
        result["size_bytes"] = size

        if size > P2AUD_C013_MAX_BYTES:
            result["read_status"] = "SKIPPED_SIZE_LIMIT"
            return result

        raw = path.read_bytes()

        result["raw"] = raw
        result["sha256"] = hashlib.sha256(raw).hexdigest()
        result["read_status"] = "READ_SUCCESS"

    except OSError as exc:
        result["read_status"] = "READ_ERROR"
        result["error_type"] = type(exc).__name__

    return result


P2AUD_C013_VARIANT_LEDGER = []
P2AUD_C013_VARIANT_FAMILY_COUNTS = Counter()
P2AUD_C013_VARIANT_SNAPSHOTS = {}

for variant in P2AUD_C011_VARIANT_REGISTRY:
    relative_path = variant["relative_path"]
    family_id = variant["notebook_prefix"] or "UNCLASSIFIED"

    snapshot = p2aud_c013_read_snapshot(relative_path)

    baseline_sha = variant["sha256"]
    current_sha = snapshot["sha256"]

    if snapshot["read_status"] != "READ_SUCCESS":
        snapshot_status = snapshot["read_status"]
    elif current_sha == baseline_sha:
        snapshot_status = "MATCH_C011"
    else:
        snapshot_status = "CHANGED_SINCE_C011"

    if variant["path_role"] == "ARCHIVE_OR_BACKUP":
        classification = "ARCHIVED_OR_BACKUP"
    elif variant["path_role"] == "AUDIT":
        classification = "INDEPENDENT_AUDIT"
    elif family_id.startswith("NB"):
        classification = "ACTIVE_LOCATION_VARIANT_CANDIDATE"
    else:
        classification = "AUXILIARY_NOTEBOOK"

    unrun = variant["saved_unrun_code_cells"]
    errors = variant["saved_error_outputs"]

    has_saved_flags = (
        isinstance(unrun, int) and unrun > 0
    ) or (
        isinstance(errors, int) and errors > 0
    )

    P2AUD_C013_VARIANT_SNAPSHOTS[relative_path] = {
        "sha256_from_c011": baseline_sha,
        "sha256_current": current_sha,
        "snapshot_status": snapshot_status,
    }

    P2AUD_C013_VARIANT_LEDGER.append({
        "relative_path": relative_path,
        "family_id": family_id,
        "path_role": variant["path_role"],
        "classification_hint": classification,
        "sha256_from_c011": baseline_sha,
        "sha256_current": current_sha,
        "snapshot_status": snapshot_status,
        "saved_unrun_code_cells": unrun,
        "saved_error_outputs": errors,
        "saved_execution_flags_from_c011": has_saved_flags,
        "current_execution_counts_reverified": False,
        "replacement_for_primary": "NOT_AUTHORIZED",
        "formal_acceptance": "NOT_EVALUATED",
    })

    P2AUD_C013_VARIANT_FAMILY_COUNTS[family_id] += 1

assert len(P2AUD_C013_VARIANT_LEDGER) == 26


# Inspect references in primary NB22/NB23 and relevant NB22 variants.
# Inspect the current observed bytes, explicitly labeling any drift.

P2AUD_C013_REFERENCE_PATTERNS = {
    "HANDOFF_CONTRACT": r"phase3_handoff_contract(?:\.json)?",
    "READINESS_REPORT": r"phase3_readiness(?:\.md)?",
    "REQUEST_TEST": r"test_phase3_request",
    "RESPONSE_TEST": r"test_phase3_response",
    "VERSION_TRACE_TEST": r"test_version_trace",
    "RELEASE_MANIFEST": r"release[_ -]?manifest",
    "FORMAL_READINESS": (
        r"formal_readiness|accepted_for_phase3|phase2_completed"
    ),
    "CONTRACT_TEST_LOGIC": (
        r"pytest|unittest|def test_|contract.tests?"
    ),
}

P2AUD_C013_COMPILED_PATTERNS = {
    label: re.compile(pattern, re.IGNORECASE)
    for label, pattern in P2AUD_C013_REFERENCE_PATTERNS.items()
}

P2AUD_C013_REFERENCE_INPUTS = []

for notebook_id in ("NB22", "NB23"):
    record = P2AUD_C011_NOTEBOOK_REGISTRY[notebook_id]

    P2AUD_C013_REFERENCE_INPUTS.append({
        "notebook_id": notebook_id,
        "relative_path": record["selected_path"],
        "expected_sha256": (
            P2AUD_C012_NOTEBOOK_EVIDENCE[notebook_id]["sha256"]
        ),
        "baseline_cell": "C012",
        "kind": "PRIMARY",
    })

for variant in P2AUD_C011_VARIANT_REGISTRY:
    if (
        variant["notebook_prefix"] == "NB22"
        and variant["path_role"] == "PHASE2_ROOT"
    ):
        P2AUD_C013_REFERENCE_INPUTS.append({
            "notebook_id": "NB22",
            "relative_path": variant["relative_path"],
            "expected_sha256": variant["sha256"],
            "baseline_cell": "C011",
            "kind": "VARIANT",
        })

P2AUD_C013_NOTEBOOK_REFERENCES = []

for reference_input in P2AUD_C013_REFERENCE_INPUTS:
    relative_path = reference_input["relative_path"]
    snapshot = p2aud_c013_read_snapshot(relative_path)

    current_sha = snapshot["sha256"]
    expected_sha = reference_input["expected_sha256"]

    if snapshot["read_status"] != "READ_SUCCESS":
        snapshot_status = snapshot["read_status"]
    elif current_sha == expected_sha:
        snapshot_status = (
            "MATCH_" + reference_input["baseline_cell"]
        )
    else:
        snapshot_status = (
            "CHANGED_SINCE_" + reference_input["baseline_cell"]
        )

    reference_result = {
        "notebook_id": reference_input["notebook_id"],
        "relative_path": relative_path,
        "kind": reference_input["kind"],
        "baseline_cell": reference_input["baseline_cell"],
        "sha256_baseline": expected_sha,
        "sha256_observed": current_sha,
        "snapshot_status": snapshot_status,
        "reference_status": "NOT_SCANNED",
        "matches": {},
        "evidence_status": (
            "STATIC_SOURCE_REFERENCE_NOT_EXECUTION_PROOF"
        ),
    }

    if snapshot["read_status"] != "READ_SUCCESS":
        reference_result["reference_status"] = (
            "SKIPPED_UNREADABLE"
        )
        P2AUD_C013_NOTEBOOK_REFERENCES.append(
            reference_result
        )
        continue

    try:
        notebook = json.loads(
            snapshot["raw"].decode("utf-8-sig")
        )

        cells = notebook.get("cells", [])

        if not isinstance(cells, list):
            raise ValueError("Notebook cells must be a list")

        matches_by_label = {
            label: {
                "line_match_count": 0,
                "examples": [],
            }
            for label in P2AUD_C013_REFERENCE_PATTERNS
        }

        for cell_number, cell in enumerate(
            cells, start=1
        ):
            if not isinstance(cell, dict):
                continue

            if cell.get("cell_type") not in {
                "code", "markdown"
            }:
                continue

            source = cell.get("source", "")

            if isinstance(source, list):
                source = "".join(
                    part for part in source
                    if isinstance(part, str)
                )

            if not isinstance(source, str):
                continue

            for line_number, line in enumerate(
                source.splitlines(), start=1
            ):
                for label, regex in (
                    P2AUD_C013_COMPILED_PATTERNS.items()
                ):
                    match = regex.search(line)

                    if not match:
                        continue

                    evidence = matches_by_label[label]
                    evidence["line_match_count"] += 1

                    if len(evidence["examples"]) < 3:
                        evidence["examples"].append({
                            "cell_number": cell_number,
                            "cell_type": cell.get("cell_type"),
                            "line_in_cell": line_number,
                            "matched_term": match.group(0)[:70],
                        })

        reference_result["matches"] = {
            label: evidence
            for label, evidence in matches_by_label.items()
            if evidence["line_match_count"] > 0
        }

        reference_result["reference_status"] = (
            "STATIC_REFERENCES_EXTRACTED"
        )

    except (ValueError, TypeError, UnicodeError) as exc:
        reference_result["reference_status"] = (
            "NOTEBOOK_PARSE_ERROR"
        )
        reference_result["error_type"] = type(exc).__name__

    P2AUD_C013_NOTEBOOK_REFERENCES.append(
        reference_result
    )


# Preserve snapshot drift as an explicit unresolved audit finding.

P2AUD_C013_VARIANT_DRIFT = [
    {
        "relative_path": record["relative_path"],
        "sha256_from_c011": record["sha256_from_c011"],
        "sha256_current": record["sha256_current"],
        "snapshot_status": record["snapshot_status"],
    }
    for record in P2AUD_C013_VARIANT_LEDGER
    if record["snapshot_status"] != "MATCH_C011"
]

P2AUD_C013_REFERENCE_DRIFT = [
    {
        "relative_path": record["relative_path"],
        "kind": record["kind"],
        "snapshot_status": record["snapshot_status"],
        "sha256_baseline": record["sha256_baseline"],
        "sha256_observed": record["sha256_observed"],
    }
    for record in P2AUD_C013_NOTEBOOK_REFERENCES
    if record["snapshot_status"] not in {
        "MATCH_C011", "MATCH_C012"
    }
]

P2AUD_C013_EXECUTION_FLAGS = [
    record
    for record in P2AUD_C013_VARIANT_LEDGER
    if record["saved_execution_flags_from_c011"]
]

P2AUD_C013_ALT_COUNTS = Counter(
    record["search_status"]
    for record in P2AUD_C013_ALTERNATIVE_LOCATIONS.values()
)

P2AUD_C013_CLASS_COUNTS = Counter(
    record["classification_hint"]
    for record in P2AUD_C013_VARIANT_LEDGER
)

P2AUD_C013_SNAPSHOT_COUNTS = Counter(
    record["snapshot_status"]
    for record in P2AUD_C013_VARIANT_LEDGER
)

P2AUD_C013_HAS_DRIFT = bool(
    P2AUD_C013_VARIANT_DRIFT
    or P2AUD_C013_REFERENCE_DRIFT
)

P2AUD_C013_REPORT = {
    "schema_version": (
        "P2AUD_ARTIFACT_AND_VARIANT_RECONCILIATION_V2"
    ),
    "cell_id": "P2AUD-C013",
    "action": 13,
    "status": (
        "P2AUD_C013_SNAPSHOT_DRIFT_REVIEW_REQUIRED"
        if P2AUD_C013_HAS_DRIFT
        else "P2AUD_C013_ALTERNATIVES_STAGED_NOT_FORMALLY_VALIDATED"
    ),
    "missing_planned_paths_in_c012": len(
        P2AUD_C013_MISSING_TARGETS
    ),
    "scan_roots": P2AUD_C013_SCAN_ROOT_STATUS,
    "files_scanned": len(P2AUD_C013_SCANNED_FILES),
    "directories_scanned": len(
        P2AUD_C013_SCANNED_DIRECTORIES
    ),
    "scan_truncated": P2AUD_C013_SCAN_TRUNCATED,
    "alternative_location_status_counts": dict(
        sorted(P2AUD_C013_ALT_COUNTS.items())
    ),
    "alternative_locations": (
        P2AUD_C013_ALTERNATIVE_LOCATIONS
    ),
    "release_named_file_candidates_count": len(
        P2AUD_C013_RELEASE_FILE_CANDIDATES
    ),
    "release_named_file_examples": (
        P2AUD_C013_RELEASE_FILE_CANDIDATES[:20]
    ),
    "release_named_directory_candidates_count": len(
        P2AUD_C013_RELEASE_DIR_CANDIDATES
    ),
    "release_named_directory_examples": (
        P2AUD_C013_RELEASE_DIR_CANDIDATES[:15]
    ),
    "variant_notebooks_accounted_for": len(
        P2AUD_C013_VARIANT_LEDGER
    ),
    "variant_family_counts": dict(
        sorted(P2AUD_C013_VARIANT_FAMILY_COUNTS.items())
    ),
    "variant_classification_counts": dict(
        sorted(P2AUD_C013_CLASS_COUNTS.items())
    ),
    "variant_snapshot_status_counts": dict(
        sorted(P2AUD_C013_SNAPSHOT_COUNTS.items())
    ),
    "variant_snapshot_drift_count": len(
        P2AUD_C013_VARIANT_DRIFT
    ),
    "variant_snapshot_drift": (
        P2AUD_C013_VARIANT_DRIFT
    ),
    "variants_with_saved_execution_flags": len(
        P2AUD_C013_EXECUTION_FLAGS
    ),
    "variant_execution_flag_examples": [
        {
            "relative_path": record["relative_path"],
            "saved_unrun_code_cells": (
                record["saved_unrun_code_cells"]
            ),
            "saved_error_outputs": (
                record["saved_error_outputs"]
            ),
            "counts_reverified": False,
        }
        for record in P2AUD_C013_EXECUTION_FLAGS[:15]
    ],
    "notebook_reference_files_examined": len(
        P2AUD_C013_NOTEBOOK_REFERENCES
    ),
    "notebook_reference_summary": [
        {
            "relative_path": record["relative_path"],
            "kind": record["kind"],
            "snapshot_status": record["snapshot_status"],
            "reference_status": record["reference_status"],
            "reference_match_counts": {
                label: evidence["line_match_count"]
                for label, evidence in record["matches"].items()
            },
        }
        for record in P2AUD_C013_NOTEBOOK_REFERENCES
    ],
    "reference_snapshot_drift_count": len(
        P2AUD_C013_REFERENCE_DRIFT
    ),
    "reference_snapshot_drift": (
        P2AUD_C013_REFERENCE_DRIFT
    ),
    "snapshot_drift_requires_reconciliation": (
        P2AUD_C013_HAS_DRIFT
    ),
    "primary_notebooks_modified_by_audit": 0,
    "variants_promoted_to_primary": 0,
    "release_formally_accepted": False,
    "requirements_implementation_accepted": 0,
    "formal_gates_passed": 0,
    "files_written": 0,
    "new_api_calls": 0,
    "git_operations": 0,
    "next_cell": "P2AUD-C014",
}

assert P2AUD_C013_REPORT[
    "variant_notebooks_accounted_for"
] == 26

assert P2AUD_C013_REPORT[
    "variants_promoted_to_primary"
] == 0

print(
    json.dumps(
        P2AUD_C013_REPORT,
        ensure_ascii=False,
        indent=2,
    )
)
````
