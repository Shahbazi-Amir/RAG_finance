# P2AUD — ممیزی جامع نقشه و اجرای واقعی فاز دوم RAG_finance

**عنوان:** Phase 2 Roadmap & Execution Independent Audit

**نوت‌بوک:** `phase2_roadmap_execution_audit.ipynb`

**مسیر:** `notebooks/phase2/audit/`

**شاخهٔ مرجع:** `rag-phase2`

**دامنه:** ممیزی نقشه‌ها، تصمیم‌ها، نوت‌بوک‌ها، شواهد، کیفیت، Release و تدوین نقشهٔ واقعی As-Built فاز دوم.

## هدف

مشخص کنیم فاز دوم طبق نقشه قرار بوده چه چیزهایی بسازد، در عمل چه چیزهایی ساخته شده، کدام شروط واقعاً پذیرفته شده‌اند و کدام تعهدات هنوز باز هستند.

## ترتیب ممیزی

1. تثبیت نقشه‌های اصلی و نسخه‌های واقعی.
2. استخراج الزامات، Gateها و خروجی‌های مورد انتظار.
3. شناسایی اصلاحیه‌ها، تصمیم‌های جایگزین و اختلاف‌های نقشه‌ها.
4. تهیهٔ ماتریس جامع الزامات فاز دوم.
5. بررسی تمام نوت‌بوک‌های NB00 تا NB23 و نسخه‌های انشعابی.
6. صحت‌سنجی Artifactها، Manifestها، گزارش‌ها و نتایج آزمون.
7. بررسی سلامت ساختاری و اجرای پاک نوت‌بوک‌های نهایی.
8. تدوین نقشهٔ واقعی اجراشده (As-Built Roadmap).
9. تهیهٔ گزارش مدیریتی و تصویری برای کارفرما.

## مراجع اولیه

- `data/Road_Maps/RAG_finance_phase2_modular_notebooks_execution_plan.md`
- `data/Road_Maps/RAG_finance_phase2_operational_roadmap.md`
- `data/Road_Maps/RAG_finance_master_roadmap.md`

ممیزی‌های تاریخی، چک‌لیست‌ها، ADRها و اصلاحیه‌ها پس از شناسایی نسخه و منشأ به دفتر شواهد افزوده می‌شوند.

## قواعد غیرقابل‌تغییر

- GitHub فقط خواندنی؛ بدون Commit، Push، ساخت شاخه یا تغییر مخزن.
- بدون اجرای API پولی، مدل، Pipeline سنگین یا بازسازی تاریخی بدون مجوز صریح.
- نبود فایل، نبود شاهد و شکست Parser به‌صورت جداگانه ثبت شوند.
- نتیجهٔ قدیمی یا گزارش ذخیره‌شده به‌تنهایی اثبات اجرای جدید نیست.
- اختلاف واقعی بین اسناد بدون تصمیم معتبر یا تأیید کاربر حل‌شده اعلام نشود.
- نسخهٔ مصوب (Planned) و نسخهٔ اجراشده (As-Built) جدا باقی بمانند.
- هر خروجی دارای هویت منبع، نسخه، هش، مسیر، Validation و محدودیت باشد.
- گزارش‌های موجود بی‌ضرورت دوباره ساخته نشوند.
- تمام Markdownها فارسی RTL و تمام سل‌های Python شماره‌گذاری‌شده باشند.

## تفکیک وضعیت‌ها

چهار وضعیت مستقل‌اند:

1. `IMPLEMENTATION_COMPLETE` — کار پیاده‌سازی شده است.
2. `TECHNICAL_VALIDATION_PASS` — آزمون فنی در دامنهٔ مشخص موفق بوده است.
3. `QUALITY_GATE_PASS` — معیار کیفیت مورد توافق احراز شده است.
4. `FORMAL_RELEASE_ACCEPTED` — Release رسمی و شرایط تحویل پذیرفته شده‌اند.

موفقیت یکی از این موارد، اثبات سه مورد دیگر نیست.

**خروجی:** گزارش JSON راه‌اندازی در خروجی سل.

**شرط پایان:** سل بدون خطا اجرا شود و هیچ فایل یا عملیات Git ایجاد نکند.

## اقدام ۲ — شناسایی نقشه‌های مرجع فاز دوم

**شناسه:** `P2AUD-C002`

### هدف

پیش از مقایسهٔ نقشه‌ها، هویت و نسخهٔ فایل‌های مرجع را از مخزن محلی شناسایی کنیم.

### ورودی‌ها

۱. نقشهٔ ماژولار فاز دوم: `RAG_finance_phase2_modular_notebooks_execution_plan.md`

۲. نقشهٔ عملیاتی فاز دوم: `RAG_finance_phase2_operational_roadmap.md`

۳. نقشهٔ کلان: `RAG_finance_master_roadmap.md`

### کنترل‌ها

- شناسایی ریشهٔ واقعی مخزن از مسیر نوت‌بوک.
- بررسی وجود و خوانایی هر سه فایل.
- محاسبهٔ SHA-256 و Git Blob SHA.
- مقایسهٔ نسخهٔ محلی با نسخهٔ مرجع قبلاً مشاهده‌شده.
- ثبت شاخه و HEAD محلی در صورت دسترسی.
- حفظ متن کامل فایل‌ها برای ممیزی مراحل بعد.

### شرط پایان

هر سه نقشه خوانده شده باشند و وضعیت تطابق نسخه‌ها صریح ثبت شود. اختلاف نسخه نباید پنهان یا خودکار اصلاح شود.

**محدودیت:** هیچ فایل، Artifact یا گزارش جدیدی در این اقدام نوشته نمی‌شود؛ هیچ Pipeline یا API اجرا نمی‌شود و هیچ تغییری در Git انجام نمی‌گیرد.

</div>


````python
# P2AUD-C002 — Read-only discovery and fingerprinting of Phase 2 roadmaps
# Action 2 | NO API | NO FILE WRITES | NO GIT WRITES

assert P2AUD_BOOTSTRAP["status"] == "P2AUD_C001_BOOTSTRAP_READY"

P2AUD_ROADMAP_SPECS = (
    {
        "id": "PHASE2_MODULAR",
        "path": "data/Road_Maps/RAG_finance_phase2_modular_notebooks_execution_plan.md",
        "reference_blob_sha": "e3ec855f54922b7e18d7541bf1aadb39d5c759d3",
    },
    {
        "id": "PHASE2_OPERATIONAL",
        "path": "data/Road_Maps/RAG_finance_phase2_operational_roadmap.md",
        "reference_blob_sha": "a8215a10695e9bed8ab57d78cac8cdc059e486d3",
    },
    {
        "id": "MASTER_ROADMAP",
        "path": "data/Road_Maps/RAG_finance_master_roadmap.md",
        "reference_blob_sha": "c239b7081e88e1f9d806a3e8095f6482a20cbf86",
    },
)

# Normally leave this as None.
# If the notebook is outside the repository, set the local root explicitly.
P2AUD_REPO_ROOT_OVERRIDE = None

if P2AUD_REPO_ROOT_OVERRIDE is None:
    candidate_roots = (
        Path.cwd(),
        *Path.cwd().parents,
    )
else:
    candidate_roots = (
        Path(P2AUD_REPO_ROOT_OVERRIDE).expanduser().resolve(),
    )

P2AUD_REPO_ROOT = None

for candidate in candidate_roots:
    if all(
        (candidate / spec["path"]).is_file()
        for spec in P2AUD_ROADMAP_SPECS
    ):
        P2AUD_REPO_ROOT = candidate.resolve()
        break

if P2AUD_REPO_ROOT is None:
    raise FileNotFoundError(
        "P2AUD_C002_REPOSITORY_ROOT_NOT_FOUND: "
        "The three required roadmap files were not found together. "
        "Check your notebook location or set "
        "P2AUD_REPO_ROOT_OVERRIDE to the local repository root."
    )

P2AUD_ROADMAP_TEXTS = {}
P2AUD_ROADMAP_INDEX = {}

for spec in P2AUD_ROADMAP_SPECS:
    source_path = P2AUD_REPO_ROOT / spec["path"]

    raw = source_path.read_bytes()
    content = raw.decode("utf-8-sig")

    if not content.strip():
        raise ValueError(
            f"Empty roadmap: {spec['id']}"
        )

    source_lines = content.splitlines()

    headings = [
        {
            "line": number,
            "text": line.strip(),
        }
        for number, line in enumerate(source_lines, start=1)
        if re.match(r"^#{1,6}\s+", line)
    ]

    if not headings:
        raise ValueError(
            f"No Markdown headings found in {spec['id']}"
        )

    sha256 = hashlib.sha256(raw).hexdigest()

    git_blob_header = (
        b"blob " + str(len(raw)).encode("ascii") + b"\0"
    )

    git_blob_sha = hashlib.sha1(
        git_blob_header + raw
    ).hexdigest()

    version_status = (
        "MATCH_REFERENCE"
        if git_blob_sha == spec["reference_blob_sha"]
        else "DIFFERS_FROM_REFERENCE"
    )

    P2AUD_ROADMAP_TEXTS[spec["id"]] = content

    P2AUD_ROADMAP_INDEX[spec["id"]] = {
        "relative_path": spec["path"],
        "size_bytes": len(raw),
        "line_count": len(source_lines),
        "heading_count": len(headings),
        "sha256": sha256,
        "git_blob_sha": git_blob_sha,
        "reference_blob_sha": spec["reference_blob_sha"],
        "version_status": version_status,
        "headings": headings,
    }

def p2aud_read_git_value(*arguments):
    try:
        result = subprocess.run(
            ["git", "-C", str(P2AUD_REPO_ROOT), *arguments],
            capture_output=True,
            text=True,
            check=False,
            timeout=8,
        )
        if result.returncode == 0:
            return result.stdout.strip() or None
    except (OSError, subprocess.TimeoutExpired):
        pass
    return None

P2AUD_LOCAL_BRANCH = p2aud_read_git_value(
    "branch", "--show-current"
)

P2AUD_LOCAL_HEAD = p2aud_read_git_value(
    "rev-parse", "HEAD"
)

version_counts = Counter(
    item["version_status"]
    for item in P2AUD_ROADMAP_INDEX.values()
)

all_reference_versions_match = (
    version_counts["MATCH_REFERENCE"]
    == len(P2AUD_ROADMAP_SPECS)
)

P2AUD_C002_REPORT = {
    "schema_version": "P2AUD_ROADMAP_SOURCE_INVENTORY_V1",
    "cell_id": "P2AUD-C002",
    "action": 2,
    "status": (
        "P2AUD_C002_ROADMAP_SOURCES_VERIFIED"
        if all_reference_versions_match
        else "P2AUD_C002_VERSION_DIFFERENCES_REQUIRE_REVIEW"
    ),
    "repository_root_name": P2AUD_REPO_ROOT.name,
    "local_branch": P2AUD_LOCAL_BRANCH,
    "local_head": P2AUD_LOCAL_HEAD,
    "branch_matches_expected": (
        P2AUD_LOCAL_BRANCH == "rag-phase2"
    ),
    "roadmaps_loaded": len(P2AUD_ROADMAP_INDEX),
    "version_counts": dict(version_counts),
    "roadmaps": {
        source_id: {
            key: value
            for key, value in item.items()
            if key != "headings"
        }
        for source_id, item in P2AUD_ROADMAP_INDEX.items()
    },
    "requirements_evaluated": 0,
    "files_written": 0,
    "new_api_calls": 0,
    "git_write_operations": 0,
    "next_cell": "P2AUD-C003",
}

assert P2AUD_C002_REPORT["roadmaps_loaded"] == 3
assert len(P2AUD_ROADMAP_TEXTS) == 3

print(
    json.dumps(
        P2AUD_C002_REPORT,
        ensure_ascii=False,
        indent=2,
    )
)
````

## اقدام ۳ — استخراج ساختار نقشه‌های فاز دوم

**شناسه:** `P2AUD-C003`

### هدف

ساخت فهرست قابل ردیابی از ساختار سه نقشهٔ مرجع و آماده‌سازی اطلاعات برای مقایسهٔ الزامات واقعی.

### ورودی‌ها

- متن نقشهٔ ماژولار فاز دوم از C002
- متن نقشهٔ عملیاتی فاز دوم از C002
- متن نقشهٔ کلان پروژه از C002

### عملیات

1. استخراج تمام عنوان‌ها با شمارهٔ خط.
2. شناسایی نوت‌بوک‌های NB00 تا NB23 در نقشهٔ ماژولار.
3. شناسایی بسته‌های P2-01 تا P2-14 در نقشهٔ عملیاتی.
4. استخراج مرز متنی هر نوت‌بوک و بسته، شامل فعالیت‌ها، خروجی‌ها و Gateهای آن.
5. ثبت بخش‌های مشترک و فراگیر نقشه‌ها برای بررسی‌های بعدی.
6. کنترل شماره‌های تکراری یا مفقود بدون اصلاح خودکار.

### شرط پایان

ساختار اسناد به‌صورت قابل بازیابی در حافظه آماده باشد و هیچ الزام یا اختلافی بدون بررسی محتوایی قضاوت نشود.

### محدودهٔ اجرا

فقط خواندن داده‌های موجود در حافظه؛ بدون درخواست API، تغییر Git، اجرای Pipeline یا ایجاد فایل.

**خروجی:** گزارش ساختاری C003 و داده‌های استخراج‌شده برای C004.

</div>


````python
# P2AUD-C003 — Extract Roadmap Structure and Section Boundaries
# Action 3 | READ_ONLY | NO_API | NO_GIT_WRITES

assert P2AUD_C002_REPORT["status"] == (
    "P2AUD_C002_ROADMAP_SOURCES_VERIFIED"
)

assert len(P2AUD_ROADMAP_TEXTS) == 3

P2AUD_C003_HEADINGS = {}
P2AUD_C003_UNITS = {
    "notebooks": {},
    "operational_packages": {},
    "master_sections": {},
}

def p2aud_extract_headings(source_id):
    lines = P2AUD_ROADMAP_TEXTS[source_id].splitlines()
    headings = []

    for number, line in enumerate(lines, start=1):
        match = re.match(
            r"^(#{1,6})[ \t]+(.+?)\s*$",
            line,
        )

        if match:
            headings.append({
                "line": number,
                "level": len(match.group(1)),
                "title": match.group(2).strip(),
            })

    return lines, headings


def p2aud_extract_unit(lines, headings, heading):
    start = heading["line"]
    end = len(lines)

    for next_heading in headings:
        if (
            next_heading["line"] > start
            and next_heading["level"] <= heading["level"]
        ):
            end = next_heading["line"] - 1
            break

    section_lines = lines[start - 1:end]

    subsections = [
        {
            "line": item["line"],
            "title": item["title"],
        }
        for item in headings
        if (
            start < item["line"] <= end
            and item["level"] > heading["level"]
        )
    ]

    candidate_items = [
        {
            "line": number,
            "text": text.strip(),
        }
        for number, text in enumerate(
            section_lines,
            start=start,
        )
        if re.match(
            r"^\s*(?:[-*]|\d+[.)])\s+\S",
            text,
        )
    ]

    raw_section = "\n".join(section_lines)

    return {
        "title": heading["title"],
        "start_line": start,
        "end_line": end,
        "line_count": len(section_lines),
        "subsections": subsections,
        "candidate_items": candidate_items,
        "section_sha256": hashlib.sha256(
            raw_section.encode("utf-8")
        ).hexdigest(),
        "raw_text": raw_section,
    }


for source_id in (
    "PHASE2_MODULAR",
    "PHASE2_OPERATIONAL",
    "MASTER_ROADMAP",
):
    lines, headings = p2aud_extract_headings(
        source_id
    )

    expected_count = P2AUD_ROADMAP_INDEX[
        source_id
    ]["heading_count"]

    assert len(headings) == expected_count, (
        f"Heading count changed: {source_id}"
    )

    P2AUD_C003_HEADINGS[source_id] = headings

    for heading in headings:
        if heading["level"] != 2:
            continue

        title = heading["title"]

        if source_id == "PHASE2_MODULAR":
            match = re.match(
                r"^Notebook\s+(\d{2})\s+[—-]",
                title,
            )

            if match:
                unit_id = "NB" + match.group(1)

                if unit_id in P2AUD_C003_UNITS[
                    "notebooks"
                ]:
                    raise ValueError(
                        f"Duplicate notebook: {unit_id}"
                    )

                P2AUD_C003_UNITS[
                    "notebooks"
                ][unit_id] = p2aud_extract_unit(
                    lines, headings, heading
                )

        elif source_id == "PHASE2_OPERATIONAL":
            match = re.search(
                r"\bP2-(\d{2})\b",
                title,
            )

            if match:
                unit_id = "P2-" + match.group(1)

                if unit_id in P2AUD_C003_UNITS[
                    "operational_packages"
                ]:
                    raise ValueError(
                        f"Duplicate package: {unit_id}"
                    )

                P2AUD_C003_UNITS[
                    "operational_packages"
                ][unit_id] = p2aud_extract_unit(
                    lines, headings, heading
                )

        elif source_id == "MASTER_ROADMAP":
            if re.match(
                r"^[۶6][.٫]\s*فاز\s*[۲2]",
                title,
            ):
                P2AUD_C003_UNITS[
                    "master_sections"
                ]["PHASE2_SCOPE"] = (
                    p2aud_extract_unit(
                        lines, headings, heading
                    )
                )

            elif (
                "قرارداد تحویل بین فازها" in title
            ):
                P2AUD_C003_UNITS[
                    "master_sections"
                ]["PHASE_HANDOFF"] = (
                    p2aud_extract_unit(
                        lines, headings, heading
                    )
                )


expected_notebooks = {
    f"NB{i:02d}" for i in range(24)
}

expected_packages = {
    f"P2-{i:02d}" for i in range(1, 15)
}

actual_notebooks = set(
    P2AUD_C003_UNITS["notebooks"]
)

actual_packages = set(
    P2AUD_C003_UNITS["operational_packages"]
)

missing_notebooks = sorted(
    expected_notebooks - actual_notebooks
)

missing_packages = sorted(
    expected_packages - actual_packages
)

unexpected_notebooks = sorted(
    actual_notebooks - expected_notebooks
)

unexpected_packages = sorted(
    actual_packages - expected_packages
)

master_section_ids = sorted(
    P2AUD_C003_UNITS["master_sections"]
)

P2AUD_C003_STRUCTURE_OK = (
    not missing_notebooks
    and not missing_packages
    and not unexpected_notebooks
    and not unexpected_packages
    and len(master_section_ids) == 2
)

candidate_line_counts = {
    "modular_notebook_items": sum(
        len(unit["candidate_items"])
        for unit in P2AUD_C003_UNITS[
            "notebooks"
        ].values()
    ),
    "operational_package_items": sum(
        len(unit["candidate_items"])
        for unit in P2AUD_C003_UNITS[
            "operational_packages"
        ].values()
    ),
}

P2AUD_C003_REPORT = {
    "schema_version": "P2AUD_ROADMAP_STRUCTURE_V1",
    "cell_id": "P2AUD-C003",
    "action": 3,
    "status": (
        "P2AUD_C003_STRUCTURE_EXTRACTED"
        if P2AUD_C003_STRUCTURE_OK
        else "P2AUD_C003_STRUCTURE_REQUIRES_REVIEW"
    ),
    "reference_versions_verified": True,
    "heading_counts": {
        source_id: len(headings)
        for source_id, headings
        in P2AUD_C003_HEADINGS.items()
    },
    "notebooks_detected": len(
        actual_notebooks
    ),
    "notebook_ids": sorted(
        actual_notebooks
    ),
    "operational_packages_detected": len(
        actual_packages
    ),
    "operational_package_ids": sorted(
        actual_packages
    ),
    "master_sections": master_section_ids,
    "missing_notebooks": missing_notebooks,
    "missing_packages": missing_packages,
    "unexpected_notebooks": unexpected_notebooks,
    "unexpected_packages": unexpected_packages,
    "candidate_line_counts": candidate_line_counts,
    "requirements_evaluated": 0,
    "conflicts_evaluated": 0,
    "scope": "STRUCTURAL_EXTRACTION_ONLY",
    "files_written": 0,
    "new_api_calls": 0,
    "git_operations": 0,
    "next_cell": "P2AUD-C004",
}

assert P2AUD_C003_STRUCTURE_OK, (
    "Roadmap structure requires review"
)

print(
    json.dumps(
        P2AUD_C003_REPORT,
        ensure_ascii=False,
        indent=2,
    )
)
````

## اقدام ۴ — ساخت دفتر نامزدهای الزامات نقشهٔ فاز دوم

**شناسهٔ سل Python:** `P2AUD-C004`

### هدف

از روی متن قفل‌شدهٔ سه نقشه، تمام خطوط نامزدِ فعالیت، خروجی، Gate، معیار و شواهد را همراه با مسیر فایل، شناسهٔ بخش، شمارهٔ خط و هش مرجع در حافظه ثبت کنیم.

### ورودی

خروجی‌های موفق سل‌های `P2AUD-C002` و `P2AUD-C003`، بدون بارگذاری مجدد داده و بدون اجرای Pipeline.

### روش

1. استخراج ردیف‌های فهرست، ردیف‌های جدول و خطوط توضیحیِ هر واحد.
2. پوشش NB00–NB23، بسته‌های P2-01 تا P2-14، دو بخش مرجع نقشهٔ کلان و بخش‌های فراگیر خارج از نوت‌بوک‌ها و بسته‌ها.
3. ثبت `candidate_id` پایدار، مسیر و هش منبع، شمارهٔ خط و نوع اولیهٔ هر مورد.
4. نگه‌داشتن موارد مبهم با وضعیت `UNREVIEWED` و `NOT_CLASSIFIED`؛ هیچ تطبیق یا تعارضی هنوز پذیرفته نمی‌شود.
5. ثبت گزارش شمارش‌ها و نمونه‌های قابل‌ردیابی در خروجی سل.

### مرز اعتبار

**تعداد نامزدها معادل تعداد الزامات قطعی نیست.** بعضی خطوط توضیحی، بعضی جدول‌ها نمونه و بعضی مفاد چندخطی هستند. محتوای بلوک‌های کد از فهرست نامزدها کنار گذاشته می‌شود، ولی متن کامل تمام بخش‌ها از C003 در حافظه باقی می‌ماند. تشخیص الزامات واقعی و اختلاف نقشه‌ها مرحلهٔ بعد است.

### شرط پایان

گزارش `P2AUD_C004_CANDIDATES_STAGED_NOT_ADJUDICATED` ایجاد شود و تمام شناسه‌ها، خط‌های متن و دسته‌بندی‌ها قابل ردیابی باشند. هیچ فایل محلی، Git یا API تغییر نکند.

</div>

````python
# P2AUD-C004 — Stage traceable roadmap requirement candidates
# Action 4 | READ_ONLY | NO_API | NO_FILE_WRITES | NO_GIT_WRITES

assert P2AUD_C003_REPORT["status"] == "P2AUD_C003_STRUCTURE_EXTRACTED"
assert len(P2AUD_C003_UNITS["notebooks"]) == 24
assert len(P2AUD_C003_UNITS["operational_packages"]) == 14
assert set(P2AUD_C003_UNITS["master_sections"]) == {"PHASE2_SCOPE", "PHASE_HANDOFF"}

# Classification is only a review hint; it is not a compliance judgment.
P2AUD_C004_SECTION_RULES = (
    ("ACCEPTANCE_GATE", ("gate", "پذیرش", "شرط پایان", "شرط ورود")),
    ("DELIVERABLE", ("خروجی", "artifact", "تحویل", "فایل")),
    ("TEST_OR_METRIC", ("معیار", "metric", "test", "آزمون", "تست")),
    ("EVIDENCE", ("شاهد", "evidence", "دانش فنی", "مستندات")),
    ("ACTIVITY", ("سلول", "کارها", "گام", "مراحل", "فعالیت", "variant")),
    ("CONSTRAINT", ("قاعده", "محدودیت", "اصل", "قوانین", "امنیت")),
)


def p2aud_c004_category(section_title):
    normalized = section_title.casefold()
    for category, signals in P2AUD_C004_SECTION_RULES:
        if any(signal in normalized for signal in signals):
            return category
    return "REVIEW_REQUIRED"


def p2aud_c004_global_sections(source_id, excluded_ids):
    lines = P2AUD_ROADMAP_TEXTS[source_id].splitlines()
    headings = P2AUD_C003_HEADINGS[source_id]
    result = {}
    for heading in headings:
        if heading["level"] != 2:
            continue
        title = heading["title"]
        if source_id == "PHASE2_MODULAR" and re.match(
            r"^Notebook\s+\d{2}\s+[—-]", title
        ):
            continue
        if source_id == "PHASE2_OPERATIONAL" and re.search(
            r"\bP2-\d{2}\b", title
        ):
            continue
        key = "GLOBAL-" + str(heading["line"])
        result[key] = p2aud_extract_unit(lines, headings, heading)
    assert not set(result).intersection(excluded_ids)
    return result


P2AUD_C004_GLOBAL_SECTIONS = {
    "PHASE2_MODULAR": p2aud_c004_global_sections(
        "PHASE2_MODULAR", set(P2AUD_C003_UNITS["notebooks"])
    ),
    "PHASE2_OPERATIONAL": p2aud_c004_global_sections(
        "PHASE2_OPERATIONAL", set(P2AUD_C003_UNITS["operational_packages"])
    ),
}

P2AUD_C004_GROUPS = (
    ("PHASE2_MODULAR", "NOTEBOOK", P2AUD_C003_UNITS["notebooks"]),
    ("PHASE2_OPERATIONAL", "PACKAGE", P2AUD_C003_UNITS["operational_packages"]),
    ("MASTER_ROADMAP", "MASTER", P2AUD_C003_UNITS["master_sections"]),
    ("PHASE2_MODULAR", "GLOBAL", P2AUD_C004_GLOBAL_SECTIONS["PHASE2_MODULAR"]),
    ("PHASE2_OPERATIONAL", "GLOBAL", P2AUD_C004_GLOBAL_SECTIONS["PHASE2_OPERATIONAL"]),
)

P2AUD_C004_CANDIDATES = []
P2AUD_C004_SECTION_LEDGER = []
P2AUD_C004_SKIPPED = Counter()
P2AUD_C004_SEEN_IDS = set()

for source_id, unit_type, units in P2AUD_C004_GROUPS:
    source_path = P2AUD_ROADMAP_INDEX[source_id]["relative_path"]
    source_blob_sha = P2AUD_ROADMAP_INDEX[source_id]["git_blob_sha"]
    source_lines = P2AUD_ROADMAP_TEXTS[source_id].splitlines()
    for unit_id, unit in sorted(
        units.items(), key=lambda pair: pair[1]["start_line"]
    ):
        active_section = unit["title"]
        active_category = "REVIEW_REQUIRED"
        in_code_fence = False
        unit_candidate_count = 0
        unit_code_line_count = 0
        unit_uncategorized_count = 0
        for line_no in range(unit["start_line"], unit["end_line"] + 1):
            raw_line = source_lines[line_no - 1]
            content = raw_line.strip()
            if not content:
                continue
            if re.match(r"^\s*(```|~~~)", raw_line):
                in_code_fence = not in_code_fence
                P2AUD_C004_SKIPPED["FENCE_MARKERS"] += 1
                continue
            if in_code_fence:
                unit_code_line_count += 1
                P2AUD_C004_SKIPPED["CODE_EXAMPLE_LINES"] += 1
                continue
            heading_match = re.match(r"^#{1,6}\s+(.+?)\s*$", raw_line)
            if heading_match:
                active_section = heading_match.group(1)
                active_category = p2aud_c004_category(active_section)
                continue
            if re.fullmatch(r"[-_*]{3,}", content):
                continue
            if content.startswith(("<div", "</div", "<table", "</table")):
                P2AUD_C004_SKIPPED["HTML_LAYOUT"] += 1
                continue
            if re.match(r"^\|\s*:?-{2,}", content):
                P2AUD_C004_SKIPPED["TABLE_SEPARATOR"] += 1
                continue

            if re.match(r"^\s*(?:[-*+]\s+|\d+[.)]\s+)", raw_line):
                record_kind = "LIST_ITEM"
            elif content.startswith("|") and content.endswith("|"):
                record_kind = "TABLE_ROW"
            else:
                record_kind = "PARAGRAPH_LINE"
                if active_category == "REVIEW_REQUIRED":
                    unit_uncategorized_count += 1

            # Keep possible explanatory lines visible; never silently discard them.
            candidate_key = (
                f"{source_id}|{unit_type}|{unit_id}|{line_no}|{raw_line}"
            )
            candidate_id = (
                "P2AUD-RC-" + hashlib.sha256(
                    candidate_key.encode("utf-8")
                ).hexdigest()[:16]
            )
            if candidate_id in P2AUD_C004_SEEN_IDS:
                raise AssertionError(f"Duplicate candidate ID: {candidate_id}")
            P2AUD_C004_SEEN_IDS.add(candidate_id)
            P2AUD_C004_CANDIDATES.append({
                "candidate_id": candidate_id,
                "source_id": source_id,
                "source_path": source_path,
                "source_blob_sha": source_blob_sha,
                "unit_type": unit_type,
                "unit_id": unit_id,
                "unit_title": unit["title"],
                "section_title": active_section,
                "category_hint": active_category,
                "record_kind": record_kind,
                "source_line": line_no,
                "text": content,
                "review_status": "UNREVIEWED",
                "requirement_status": "NOT_CLASSIFIED",
                "conflict_status": "NOT_EVALUATED",
            })
            unit_candidate_count += 1
        if in_code_fence:
            raise ValueError(
                f"Unclosed Markdown fence in {source_id}/{unit_id}"
            )
        P2AUD_C004_SECTION_LEDGER.append({
            "source_id": source_id,
            "unit_type": unit_type,
            "unit_id": unit_id,
            "start_line": unit["start_line"],
            "end_line": unit["end_line"],
            "section_sha256": unit["section_sha256"],
            "candidate_count": unit_candidate_count,
            "code_example_lines_excluded": unit_code_line_count,
            "uncategorized_paragraph_lines": unit_uncategorized_count,
        })

# Verify that every staged item traces to the immutable source inventory.
for record in P2AUD_C004_CANDIDATES:
    source_line = P2AUD_ROADMAP_TEXTS[
        record["source_id"]
    ].splitlines()[record["source_line"] - 1].strip()
    assert source_line == record["text"], record["candidate_id"]

P2AUD_C004_BY_KIND = Counter(
    item["record_kind"] for item in P2AUD_C004_CANDIDATES
)
P2AUD_C004_BY_CATEGORY = Counter(
    item["category_hint"] for item in P2AUD_C004_CANDIDATES
)
P2AUD_C004_BY_SOURCE = Counter(
    item["source_id"] for item in P2AUD_C004_CANDIDATES
)
P2AUD_C004_BY_UNIT_TYPE = Counter(
    item["unit_type"] for item in P2AUD_C004_CANDIDATES
)

assert len(P2AUD_C004_CANDIDATES) == len(P2AUD_C004_SEEN_IDS)
assert sum(P2AUD_C004_BY_SOURCE.values()) == len(P2AUD_C004_CANDIDATES)
assert len(P2AUD_C004_SECTION_LEDGER) >= 40
assert all(
    item["review_status"] == "UNREVIEWED"
    for item in P2AUD_C004_CANDIDATES
)

P2AUD_C004_SAMPLE = []
for sample_type in ("NOTEBOOK", "PACKAGE", "MASTER", "GLOBAL"):
    example = next(
        (x for x in P2AUD_C004_CANDIDATES
         if x["unit_type"] == sample_type),
        None,
    )
    if example:
        P2AUD_C004_SAMPLE.append({
            "candidate_id": example["candidate_id"],
            "source_id": example["source_id"],
            "unit_id": example["unit_id"],
            "source_line": example["source_line"],
            "category_hint": example["category_hint"],
            "text_preview": example["text"][:135],
        })

P2AUD_C004_REPORT = {
    "schema_version": "P2AUD_ROADMAP_CANDIDATE_INVENTORY_V1",
    "cell_id": "P2AUD-C004",
    "action": 4,
    "status": "P2AUD_C004_CANDIDATES_STAGED_NOT_ADJUDICATED",
    "source_versions_match_reference": True,
    "units_scanned": len(P2AUD_C004_SECTION_LEDGER),
    "global_sections_scanned": sum(
        len(parts) for parts in P2AUD_C004_GLOBAL_SECTIONS.values()
    ),
    "candidate_count": len(P2AUD_C004_CANDIDATES),
    "counts_by_source": dict(sorted(P2AUD_C004_BY_SOURCE.items())),
    "counts_by_unit_type": dict(sorted(P2AUD_C004_BY_UNIT_TYPE.items())),
    "counts_by_record_kind": dict(sorted(P2AUD_C004_BY_KIND.items())),
    "counts_by_category_hint": dict(sorted(P2AUD_C004_BY_CATEGORY.items())),
    "skipped_non_candidate_lines": dict(sorted(P2AUD_C004_SKIPPED.items())),
    "sample_candidates": P2AUD_C004_SAMPLE,
    "requirements_evaluated": 0,
    "conflicts_evaluated": 0,
    "files_written": 0,
    "new_api_calls": 0,
    "git_operations": 0,
    "next_cell": "P2AUD-C005",
}

print(json.dumps(P2AUD_C004_REPORT, ensure_ascii=False, indent=2))
````

## اقدام ۵ — تفکیک فعالیت‌ها، خروجی‌ها و Gateهای نقشه‌ها

**شناسه:** `P2AUD-C005`

### هدف

ساخت دفتر کنترل تعهدات فاز دوم برای آماده‌سازی تطبیق نقشهٔ ماژولار، نقشهٔ عملیاتی و نقشهٔ کلان.

### ورودی‌ها

- ساختار معتبر نقشه‌ها از C003
- تعداد ۶۵۴ نامزد استخراج‌شده از C004
- متن کامل سه نقشه با نسخه و هش تثبیت‌شده در C002

### عملیات

۱. شناسایی عنوان‌های مربوط به فعالیت، خروجی، آزمون، شواهد، محدودیت و Gate.

۲. ثبت محدودهٔ متنی هر عنوان همراه با شمارهٔ خط و نام سند.

۳. اتصال نامزدهای استخراج‌شده به بخش‌های مربوط.

۴. مشخص‌کردن نوت‌بوک‌ها و بسته‌هایی که عنوان Gate مستقل ندارند.

۵. ثبت موارد نیازمند بررسی، بدون اعلام نقص یا شکست.

۶. ایجاد دفتر کنترل قابل‌ردیابی برای تطبیق محتوایی نقشه‌ها در اقدام بعدی.

### قواعد ممیزی

- نداشتن عنوان مستقل Gate لزوماً به معنی نداشتن شرط پذیرش نیست.
- وجود عنوان Gate به معنی اجرای موفق آن نیست.
- طبقه‌بندی عنوان‌ها صرفاً راهنمای بررسی است.
- تعارض‌ها در این اقدام هنوز قضاوت نمی‌شوند.
- هیچ عدد یا نتیجه‌ای از روی نام بخش‌ها ساخته نمی‌شود.

### خروجی

- `P2AUD_C005_ANCHORS`
- `P2AUD_C005_UNIT_LEDGER`
- `P2AUD_C005_REVIEW_FLAGS`
- `P2AUD_C005_REPORT`

### شرط پایان

تمام واحدهای استخراج‌شده در C004 بررسی ساختاری شوند و وضعیت بخش‌های نیازمند بررسی روشن باشد.

بدون ایجاد فایل، تغییر Git، فراخوانی API یا اجرای Pipeline.

</div>

````python
# P2AUD-C005 — Index Roadmap Gates, Deliverables and Evidence
# Action 5 | READ_ONLY | NO_API | NO_FILE_WRITES | NO_GIT_WRITES

assert P2AUD_C004_REPORT["status"] == (
    "P2AUD_C004_CANDIDATES_STAGED_NOT_ADJUDICATED"
)
assert len(P2AUD_C004_CANDIDATES) == 654

P2AUD_C005_ROLE_RULES = {
    "ACCEPTANCE_GATE": (
        r"\bgate\b",
        r"پذیرش",
        r"تعریف\s+done",
        r"شرط\s+پایان",
        r"شرط\s+ورود",
    ),
    "DELIVERABLE": (
        r"خروجی",
        r"artifact",
        r"deliverable",
        r"تحویل",
    ),
    "TEST_OR_METRIC": (
        r"معیار",
        r"metric",
        r"benchmark",
        r"آزمون",
        r"تست",
        r"ارزیابی",
        r"regression",
    ),
    "EVIDENCE": (
        r"شاهد",
        r"evidence",
        r"مستندات",
        r"دانش\s+فنی",
    ),
    "ACTIVITY": (
        r"سلول",
        r"فعالیت",
        r"کارها",
        r"گام",
        r"مراحل",
        r"ترتیب\s+اجرا",
    ),
    "CONSTRAINT": (
        r"قاعده",
        r"محدودیت",
        r"اصول",
        r"قوانین",
    ),
}


def p2aud_c005_heading_roles(title):
    normalized = unicodedata.normalize(
        "NFKC", title
    ).casefold()

    return [
        role
        for role, expressions in P2AUD_C005_ROLE_RULES.items()
        if any(
            re.search(expression, normalized, re.IGNORECASE)
            for expression in expressions
        )
    ]


P2AUD_C005_CANDIDATES_BY_UNIT = defaultdict(list)

for candidate in P2AUD_C004_CANDIDATES:
    key = (
        candidate["source_id"],
        candidate["unit_type"],
        candidate["unit_id"],
    )
    P2AUD_C005_CANDIDATES_BY_UNIT[key].append(candidate)

P2AUD_C005_ANCHORS = []
P2AUD_C005_UNIT_LEDGER = []
P2AUD_C005_REVIEW_FLAGS = []

for source_id, unit_type, units in P2AUD_C004_GROUPS:
    headings = P2AUD_C003_HEADINGS[source_id]
    source_info = P2AUD_ROADMAP_INDEX[source_id]

    for unit_id, unit in sorted(
        units.items(),
        key=lambda item: item[1]["start_line"],
    ):
        start = unit["start_line"]
        end = unit["end_line"]

        unit_headings = [
            heading
            for heading in headings
            if start <= heading["line"] <= end
        ]

        unit_candidates = P2AUD_C005_CANDIDATES_BY_UNIT.get(
            (source_id, unit_type, unit_id),
            [],
        )

        unit_anchors = []
        role_counter = Counter()

        for heading in unit_headings:
            roles = p2aud_c005_heading_roles(
                heading["title"]
            )

            if not roles:
                continue

            heading_start = heading["line"]
            heading_end = end

            for next_heading in unit_headings:
                if (
                    next_heading["line"] > heading_start
                    and next_heading["level"] <= heading["level"]
                ):
                    heading_end = next_heading["line"] - 1
                    break

            related_candidates = [
                item
                for item in unit_candidates
                if (
                    heading_start
                    <= item["source_line"]
                    <= heading_end
                )
            ]

            identity = (
                f"{source_id}|{unit_type}|{unit_id}|"
                f"{heading_start}|{heading['title']}|"
                f"{source_info['git_blob_sha']}"
            )

            anchor_id = (
                "P2AUD-AN-"
                + hashlib.sha256(
                    identity.encode("utf-8")
                ).hexdigest()[:16]
            )

            anchor = {
                "anchor_id": anchor_id,
                "source_id": source_id,
                "source_path": source_info["relative_path"],
                "source_blob_sha": source_info["git_blob_sha"],
                "unit_type": unit_type,
                "unit_id": unit_id,
                "heading": heading["title"],
                "heading_level": heading["level"],
                "start_line": heading_start,
                "end_line": heading_end,
                "role_hints": roles,
                "related_candidate_count": len(
                    related_candidates
                ),
                "related_candidate_ids": [
                    item["candidate_id"]
                    for item in related_candidates
                ],
                "examples": [
                    {
                        "line": item["source_line"],
                        "text": item["text"][:160],
                    }
                    for item in related_candidates[:2]
                ],
                "review_status": "UNREVIEWED",
                "gate_result": "NOT_EVALUATED",
            }

            P2AUD_C005_ANCHORS.append(anchor)
            unit_anchors.append(anchor)

            for role in roles:
                role_counter[role] += 1

        if (
            unit_type in ("NOTEBOOK", "PACKAGE")
            and role_counter["ACCEPTANCE_GATE"] == 0
        ):
            P2AUD_C005_REVIEW_FLAGS.append({
                "source_id": source_id,
                "unit_type": unit_type,
                "unit_id": unit_id,
                "flag": "NO_EXPLICIT_ACCEPTANCE_HEADING",
                "interpretation": (
                    "REVIEW_REQUIRED_NOT_A_GATE_FAILURE"
                ),
            })

        P2AUD_C005_UNIT_LEDGER.append({
            "source_id": source_id,
            "unit_type": unit_type,
            "unit_id": unit_id,
            "unit_title": unit["title"],
            "start_line": start,
            "end_line": end,
            "candidate_count": len(unit_candidates),
            "anchor_count": len(unit_anchors),
            "anchor_roles": dict(sorted(
                role_counter.items()
            )),
            "anchor_ids": [
                anchor["anchor_id"]
                for anchor in unit_anchors
            ],
            "status": "STRUCTURE_INDEXED_NOT_ADJUDICATED",
        })

anchor_ids = [
    item["anchor_id"]
    for item in P2AUD_C005_ANCHORS
]

assert len(anchor_ids) == len(set(anchor_ids)), (
    "Duplicate anchor identifiers detected"
)

assert len(P2AUD_C005_UNIT_LEDGER) == len(
    P2AUD_C004_SECTION_LEDGER
), "Unit coverage differs from C004"

assert sum(
    item["candidate_count"]
    for item in P2AUD_C005_UNIT_LEDGER
) == len(P2AUD_C004_CANDIDATES), (
    "Candidate coverage mismatch"
)

P2AUD_C005_ROLE_COUNTS = Counter()

for anchor in P2AUD_C005_ANCHORS:
    for role in anchor["role_hints"]:
        P2AUD_C005_ROLE_COUNTS[role] += 1

P2AUD_C005_SOURCE_COUNTS = Counter(
    anchor["source_id"]
    for anchor in P2AUD_C005_ANCHORS
)

P2AUD_C005_NO_GATE_BY_TYPE = Counter(
    flag["unit_type"]
    for flag in P2AUD_C005_REVIEW_FLAGS
)

P2AUD_C005_GATE_SAMPLES = [
    {
        "source_id": anchor["source_id"],
        "unit_id": anchor["unit_id"],
        "heading": anchor["heading"],
        "start_line": anchor["start_line"],
        "end_line": anchor["end_line"],
        "candidate_count": anchor["related_candidate_count"],
        "example": (
            anchor["examples"][0]
            if anchor["examples"]
            else None
        ),
    }
    for anchor in P2AUD_C005_ANCHORS
    if "ACCEPTANCE_GATE" in anchor["role_hints"]
][:8]

P2AUD_C005_REPORT = {
    "schema_version": "P2AUD_ROADMAP_CONTROL_ANCHORS_V1",
    "cell_id": "P2AUD-C005",
    "action": 5,
    "status": (
        "P2AUD_C005_CONTROL_ANCHORS_INDEXED_NOT_ADJUDICATED"
    ),
    "roadmaps_loaded": len(P2AUD_ROADMAP_TEXTS),
    "units_indexed": len(P2AUD_C005_UNIT_LEDGER),
    "candidates_preserved": len(P2AUD_C004_CANDIDATES),
    "anchors_detected": len(P2AUD_C005_ANCHORS),
    "anchor_counts_by_source": dict(sorted(
        P2AUD_C005_SOURCE_COUNTS.items()
    )),
    "anchor_counts_by_role": dict(sorted(
        P2AUD_C005_ROLE_COUNTS.items()
    )),
    "units_without_explicit_gate_heading": len(
        P2AUD_C005_REVIEW_FLAGS
    ),
    "no_gate_heading_by_unit_type": dict(sorted(
        P2AUD_C005_NO_GATE_BY_TYPE.items()
    )),
    "review_flags": P2AUD_C005_REVIEW_FLAGS,
    "sample_acceptance_anchors": P2AUD_C005_GATE_SAMPLES,
    "requirements_evaluated": 0,
    "conflicts_evaluated": 0,
    "formal_gates_passed": 0,
    "files_written": 0,
    "new_api_calls": 0,
    "git_operations": 0,
    "next_cell": "P2AUD-C006",
}

assert P2AUD_C005_REPORT["units_indexed"] == 64
assert P2AUD_C005_REPORT["candidates_preserved"] == 654
assert P2AUD_C005_REPORT["formal_gates_passed"] == 0

print(
    json.dumps(
        P2AUD_C005_REPORT,
        ensure_ascii=False,
        indent=2,
    )
)
````

## اقدام ۶ — نگاشت ۱۴ بستهٔ عملیاتی به ۲۴ نوت‌بوک

**شناسه:** `P2AUD-C006`

### هدف

بررسی رابطهٔ واقعی بین نقشهٔ عملیاتی و نقشهٔ ماژولار فاز دوم، با حفظ تمام تفاوت‌ها و شواهد.

### ورودی‌ها

- سه نقشهٔ نسخه‌دار از C002
- ساختار ۲۴ نوت‌بوک و ۱۴ بسته از C003
- دفتر ۶۵۴ نامزد الزام از C004
- دفتر ۹۹ بخش کنترلی از C005

### عملیات

````python
# P2AUD-C006 — Build provisional operational-to-notebook crosswalk
# Action 6 | READ_ONLY | NO_API | NO_FILE_WRITES | NO_GIT_WRITES

assert P2AUD_C005_REPORT["status"] == (
    "P2AUD_C005_CONTROL_ANCHORS_INDEXED_NOT_ADJUDICATED"
)
assert P2AUD_C005_REPORT["units_indexed"] == 64
assert P2AUD_C005_REPORT["candidates_preserved"] == 654

# Interpretive mapping proposal based on roadmap responsibilities.
# PRIMARY does not mean accepted implementation or complete coverage.
# SUPPORT does not mean that the operational requirement is satisfied.

P2AUD_C006_MAPPING_SPEC = {
    "P2-01": {
        "primary": ["NB00"],
        "support": [],
        "reason": "Phase 1 baseline and environment freeze",
    },
    "P2-02": {
        "primary": ["NB07"],
        "support": ["NB02", "NB22"],
        "reason": "Common schema, IDs, lineage and version contracts",
    },
    "P2-03": {
        "primary": ["NB01", "NB02"],
        "support": [],
        "reason": "Website ingestion and independent acceptance",
    },
    "P2-04": {
        "primary": ["NB03", "NB04"],
        "support": [],
        "reason": "Book inventory and ingestion",
    },
    "P2-05": {
        "primary": ["NB05", "NB06"],
        "support": [],
        "reason": "Transcript inventory and ingestion",
    },
    "P2-06": {
        "primary": ["NB07", "NB09"],
        "support": ["NB02", "NB04", "NB06", "NB22"],
        "reason": "Canonical dataset, deduplication and release evidence",
    },
    "P2-07": {
        "primary": ["NB08", "NB11"],
        "support": ["NB18"],
        "reason": "Persian normalization and source-aware chunking",
    },
    "P2-08": {
        "primary": ["NB12"],
        "support": ["NB21"],
        "reason": "Golden dataset construction and evaluation usage",
    },
    "P2-09": {
        "primary": ["NB13", "NB14", "NB15"],
        "support": ["NB22"],
        "reason": "Embedding benchmark and versioned dense/sparse indexes",
    },
    "P2-10": {
        "primary": ["NB16", "NB17", "NB21"],
        "support": ["NB14", "NB15"],
        "reason": "Retrieval, reranking and comparative evaluation",
    },
    "P2-11": {
        "primary": ["NB18", "NB19"],
        "support": ["NB20"],
        "reason": "Context, generation and claim-level citation",
    },
    "P2-12": {
        "primary": ["NB20"],
        "support": ["NB21"],
        "reason": "Answerability, safety and confidence evaluation",
    },
    "P2-13": {
        "primary": ["NB22"],
        "support": ["NB00", "NB21", "NB23"],
        "reason": "Cross-cutting tests, reproducibility and release",
    },
    "P2-14": {
        "primary": ["NB22"],
        "support": ["NB10", "NB11", "NB13", "NB19", "NB20", "NB21"],
        "reason": "Cross-cutting technical documentation and evidence",
    },
}

P2AUD_C006_EXPECTED_PACKAGES = {
    f"P2-{i:02d}" for i in range(1, 15)
}

P2AUD_C006_EXPECTED_NOTEBOOKS = {
    f"NB{i:02d}" for i in range(24)
}

assert set(P2AUD_C006_MAPPING_SPEC) == (
    P2AUD_C006_EXPECTED_PACKAGES
)

P2AUD_C006_CROSSWALK = []
P2AUD_C006_PRIMARY_COVERAGE = Counter()
P2AUD_C006_ANY_COVERAGE = Counter()

operational_units = P2AUD_C003_UNITS[
    "operational_packages"
]
modular_units = P2AUD_C003_UNITS[
    "notebooks"
]

for package_id in sorted(P2AUD_C006_MAPPING_SPEC):
    spec = P2AUD_C006_MAPPING_SPEC[package_id]
    primary = list(spec["primary"])
    support = list(spec["support"])

    assert primary, f"No primary candidate: {package_id}"
    assert len(primary) == len(set(primary))
    assert len(support) == len(set(support))
    assert not set(primary).intersection(support)

    for nb_id in primary + support:
        assert nb_id in modular_units, (
            f"Unknown notebook: {nb_id}"
        )
        P2AUD_C006_ANY_COVERAGE[nb_id] += 1

    for nb_id in primary:
        P2AUD_C006_PRIMARY_COVERAGE[nb_id] += 1

    package = operational_units[package_id]

    notebook_references = []

    for role, notebook_ids in (
        ("PRIMARY_CANDIDATE", primary),
        ("SUPPORT_CANDIDATE", support),
    ):
        for nb_id in notebook_ids:
            notebook = modular_units[nb_id]

            notebook_references.append({
                "notebook_id": nb_id,
                "role": role,
                "title": notebook["title"],
                "start_line": notebook["start_line"],
                "end_line": notebook["end_line"],
                "section_sha256": notebook["section_sha256"],
            })

    P2AUD_C006_CROSSWALK.append({
        "package_id": package_id,
        "package_title": package["title"],
        "operational_source": (
            P2AUD_ROADMAP_INDEX[
                "PHASE2_OPERATIONAL"
            ]["relative_path"]
        ),
        "operational_blob_sha": (
            P2AUD_ROADMAP_INDEX[
                "PHASE2_OPERATIONAL"
            ]["git_blob_sha"]
        ),
        "package_start_line": package["start_line"],
        "package_end_line": package["end_line"],
        "package_section_sha256": package["section_sha256"],
        "modular_source": (
            P2AUD_ROADMAP_INDEX[
                "PHASE2_MODULAR"
            ]["relative_path"]
        ),
        "modular_blob_sha": (
            P2AUD_ROADMAP_INDEX[
                "PHASE2_MODULAR"
            ]["git_blob_sha"]
        ),
        "primary_notebooks": primary,
        "support_notebooks": support,
        "notebook_references": notebook_references,
        "mapping_reason": spec["reason"],
        "mapping_status": "PROPOSED_NOT_ADJUDICATED",
        "requirement_coverage": "NOT_EVALUATED",
        "gate_result": "NOT_EVALUATED",
    })


def p2aud_c006_find_evidence(source_id, pattern):
    lines = P2AUD_ROADMAP_TEXTS[source_id].splitlines()

    for line_no, line in enumerate(lines, start=1):
        if re.search(pattern, line, flags=re.IGNORECASE):
            return {
                "source_id": source_id,
                "source_path": (
                    P2AUD_ROADMAP_INDEX[source_id][
                        "relative_path"
                    ]
                ),
                "source_blob_sha": (
                    P2AUD_ROADMAP_INDEX[source_id][
                        "git_blob_sha"
                    ]
                ),
                "line": line_no,
                "text": line.strip(),
            }

    raise ValueError(
        f"Required evidence not found: {source_id} / {pattern}"
    )


P2AUD_C006_REVIEW_SPECS = [
    {
        "review_id": "P2AUD-RV-001",
        "topic": "CITATION_COVERAGE_TARGET",
        "review_type": "METRIC_SCOPE_OR_THRESHOLD",
        "description": (
            "Modular claim citation coverage is 0.90, while "
            "operational citation coverage is 95% of factual claims. "
            "Metric scope and accepted threshold require reconciliation."
        ),
        "evidence_queries": [
            ("PHASE2_MODULAR", r"Claim citation coverage"),
            ("PHASE2_OPERATIONAL", r"Citation coverage</span>"),
        ],
    },
    {
        "review_id": "P2AUD-RV-002",
        "topic": "CONTRACT_EXECUTION_ORDER",
        "review_type": "SEQUENCE_DIFFERENCE",
        "description": (
            "Operational plan places data contracts before "
            "source pipelines; the modular flow locates unified "
            "schema NB07 after ingestion NB01-NB06."
        ),
        "evidence_queries": [
            ("PHASE2_OPERATIONAL", r"2\.\s*P2-02 Contracts"),
            ("PHASE2_OPERATIONAL", r"3\.\s*P2-03"),
            ("PHASE2_MODULAR", r"F --> G\[06"),
            ("PHASE2_MODULAR", r"G --> H\[07"),
        ],
    },
    {
        "review_id": "P2AUD-RV-003",
        "topic": "GOLDEN_CHUNKING_SEQUENCE",
        "review_type": "SEQUENCE_DIFFERENCE",
        "description": (
            "Operational plan allows Golden Dataset work "
            "concurrently with chunking, while modular flow "
            "shows NB11 followed by NB12."
        ),
        "evidence_queries": [
            ("PHASE2_OPERATIONAL", r"P2-08 Golden Dataset هم‌زمان"),
            ("PHASE2_MODULAR", r"K --> L\[11"),
            ("PHASE2_MODULAR", r"L --> M\[12"),
        ],
    },
    {
        "review_id": "P2AUD-RV-004",
        "topic": "TAXONOMY_PACKAGE_OWNERSHIP",
        "review_type": "SCOPE_COVERAGE_REVIEW",
        "description": (
            "NB10 is a dedicated taxonomy and knowledge "
            "engineering notebook, but no dedicated operational "
            "P2 package has been mapped as its primary owner. "
            "Documentation support is not functional coverage."
        ),
        "evidence_queries": [
            ("PHASE2_MODULAR", r"^## Notebook 10\s"),
            ("PHASE2_OPERATIONAL", r"^## ۱۷\. بسته P2-14"),
        ],
    },
    {
        "review_id": "P2AUD-RV-005",
        "topic": "PHASE3_HANDOFF_PACKAGE_OWNERSHIP",
        "review_type": "GLOBAL_REQUIREMENT_REVIEW",
        "description": (
            "NB23 has no dedicated operational package, "
            "but the operational roadmap includes final "
            "Phase 3 handoff requirements outside P2-01..14."
        ),
        "evidence_queries": [
            ("PHASE2_MODULAR", r"^## Notebook 23\s"),
            ("PHASE2_OPERATIONAL", r"^## ۱۹\. Gate نهایی"),
        ],
    },
]

P2AUD_C006_REVIEW_ITEMS = []

for spec in P2AUD_C006_REVIEW_SPECS:
    evidence = [
        p2aud_c006_find_evidence(source_id, pattern)
        for source_id, pattern in spec["evidence_queries"]
    ]

    P2AUD_C006_REVIEW_ITEMS.append({
        "review_id": spec["review_id"],
        "topic": spec["topic"],
        "review_type": spec["review_type"],
        "description": spec["description"],
        "evidence": evidence,
        "review_status": "REQUIRES_MANUAL_REVIEW",
        "conflict_decision": "NOT_ADJUDICATED",
        "approved_resolution": None,
    })

P2AUD_C006_UNMAPPED_PRIMARY = sorted(
    P2AUD_C006_EXPECTED_NOTEBOOKS
    - set(P2AUD_C006_PRIMARY_COVERAGE)
)

P2AUD_C006_UNMAPPED_ANY = sorted(
    P2AUD_C006_EXPECTED_NOTEBOOKS
    - set(P2AUD_C006_ANY_COVERAGE)
)

P2AUD_C006_MISSING_GATE_HEADINGS = sorted(
    flag["unit_id"]
    for flag in P2AUD_C005_REVIEW_FLAGS
    if flag["unit_type"] == "PACKAGE"
)

# Missing a separately named Gate is a structural observation,
# not an assertion that the package lacks acceptance conditions.
assert P2AUD_C006_MISSING_GATE_HEADINGS == [
    "P2-06", "P2-14"
]

assert len(P2AUD_C006_CROSSWALK) == 14
assert len(P2AUD_C006_REVIEW_ITEMS) == 5

assert all(
    item["mapping_status"] == "PROPOSED_NOT_ADJUDICATED"
    for item in P2AUD_C006_CROSSWALK
)

assert all(
    item["conflict_decision"] == "NOT_ADJUDICATED"
    for item in P2AUD_C006_REVIEW_ITEMS
)

P2AUD_C006_CROSSWALK_HASH = hashlib.sha256(
    json.dumps(
        P2AUD_C006_CROSSWALK,
        ensure_ascii=False,
        sort_keys=True,
        separators=(",", ":"),
    ).encode("utf-8")
).hexdigest()

P2AUD_C006_REPORT = {
    "schema_version": "P2AUD_ROADMAP_CROSSWALK_V1",
    "cell_id": "P2AUD-C006",
    "action": 6,
    "status": (
        "P2AUD_C006_CROSSWALK_STAGED_REVIEW_REQUIRED"
    ),
    "packages_mapped_provisionally": len(
        P2AUD_C006_CROSSWALK
    ),
    "expected_packages": 14,
    "expected_notebooks": 24,
    "primary_notebooks_covered": len(
        P2AUD_C006_PRIMARY_COVERAGE
    ),
    "notebooks_without_primary_package": (
        P2AUD_C006_UNMAPPED_PRIMARY
    ),
    "notebooks_without_any_proposed_mapping": (
        P2AUD_C006_UNMAPPED_ANY
    ),
    "packages_without_explicit_gate_heading": (
        P2AUD_C006_MISSING_GATE_HEADINGS
    ),
    "crosswalk_sha256": P2AUD_C006_CROSSWALK_HASH,
    "crosswalk_summary": [
        {
            "package": item["package_id"],
            "primary": item["primary_notebooks"],
            "support": item["support_notebooks"],
            "status": item["mapping_status"],
        }
        for item in P2AUD_C006_CROSSWALK
    ],
    "review_items": P2AUD_C006_REVIEW_ITEMS,
    "potential_differences_flagged": len(
        P2AUD_C006_REVIEW_ITEMS
    ),
    "requirements_evaluated": 0,
    "conflicts_evaluated": 0,
    "formal_gates_passed": 0,
    "files_written": 0,
    "new_api_calls": 0,
    "git_operations": 0,
    "next_cell": "P2AUD-C007",
}

print(
    json.dumps(
        P2AUD_C006_REPORT,
        ensure_ascii=False,
        indent=2,
    )
)
````

````python
# P2AUD-C007 — Verify roadmap differences and collect historical evidence
# Action 7 | READ_ONLY | NO_API | NO_FILE_WRITES | NO_GIT_WRITES

assert P2AUD_C006_REPORT["status"] == (
    "P2AUD_C006_CROSSWALK_STAGED_REVIEW_REQUIRED"
)
assert len(P2AUD_C006_REVIEW_ITEMS) == 5
assert P2AUD_C006_REPORT["conflicts_evaluated"] == 0

P2AUD_C007_HISTORY_PATHS = (
    "data/Road_Maps/RAG_finance_independent_audit_2026-08-20 .md",
    "data/Road_Maps/RAG_finance_questions_assessments_and_validation_framework_fa.md",
    "notebooks/data_load_read/ورد تحلیل/RAG_Finance_Phase2_Employer_Report_Source_FA.md",
    "notebooks/phase2/checklist/NB21_remaining_work_checklist_fa.md",
    "notebooks/phase2/checklist/NB22A_remaining_work_checklist_fa.md",
    "notebooks/phase2/checklist/NB23_phase3_handoff_checklist_fa_UPDATED.md",
)

P2AUD_C007_SEARCH_PATTERNS = {
    "P2AUD-RV-001": (
        r"citation.{0,40}(coverage|correctness)",
        r"پوشش.{0,30}استناد",
        r"claim.{0,30}citation",
        r"0[.,]9[05]",
        r"۹۰\s*٪|۹۵\s*٪",
    ),
    "P2AUD-RV-002": (
        r"data.{0,15}contract",
        r"common.{0,15}schema",
        r"unified.{0,15}schema",
        r"NB07|P2-02",
        r"قرارداد.{0,25}داده",
    ),
    "P2AUD-RV-003": (
        r"golden.{0,35}chunk",
        r"chunk.{0,35}golden",
        r"NB11|NB12",
        r"golden.dataset",
        r"قطعه.?بندی",
    ),
    "P2AUD-RV-004": (
        r"taxonomy",
        r"competency",
        r"NB10",
        r"مهندسی دانش",
    ),
    "P2AUD-RV-005": (
        r"handoff",
        r"NB23",
        r"readiness",
        r"phase.?3",
        r"تحویل.{0,20}فاز.{0,5}[۳3]",
    ),
}

P2AUD_C007_REVIEW_HINTS = {
    "P2AUD-RV-001": "POTENTIAL_THRESHOLD_CONFLICT",
    "P2AUD-RV-002": "POTENTIAL_SEQUENCE_DIFFERENCE",
    "P2AUD-RV-003": "PARALLEL_VS_SEQUENTIAL_PLAN",
    "P2AUD-RV-004": "PACKAGE_OWNERSHIP_REVIEW",
    "P2AUD-RV-005": "GLOBAL_GATE_VS_PACKAGE_OWNERSHIP",
}

P2AUD_C007_HISTORICAL_INVENTORY = []
P2AUD_C007_HISTORICAL_TEXTS = {}
P2AUD_C007_MAX_BYTES = 2_000_000

for relative_path in P2AUD_C007_HISTORY_PATHS:
    absolute_path = P2AUD_REPO_ROOT / relative_path

    record = {
        "relative_path": relative_path,
        "status": "NOT_FOUND",
        "sha256": None,
        "size_bytes": None,
        "line_count": None,
    }

    if absolute_path.is_file():
        try:
            size = absolute_path.stat().st_size

            if size > P2AUD_C007_MAX_BYTES:
                record["status"] = "SKIPPED_SIZE_LIMIT"
                record["size_bytes"] = size
            else:
                raw = absolute_path.read_bytes()
                content = raw.decode("utf-8-sig")

                record.update({
                    "status": "READ_SUCCESS",
                    "sha256": hashlib.sha256(raw).hexdigest(),
                    "size_bytes": len(raw),
                    "line_count": len(content.splitlines()),
                })

                P2AUD_C007_HISTORICAL_TEXTS[
                    relative_path
                ] = content

        except (OSError, UnicodeError) as exc:
            record["status"] = "READ_ERROR"
            record["error_type"] = type(exc).__name__

    P2AUD_C007_HISTORICAL_INVENTORY.append(record)


def p2aud_c007_primary_evidence(evidence, context_radius=2):
    source_id = evidence["source_id"]
    source_lines = P2AUD_ROADMAP_TEXTS[source_id].splitlines()
    line_no = evidence["line"]

    assert 1 <= line_no <= len(source_lines), (
        f"Invalid source line: {source_id}:{line_no}"
    )

    actual_text = source_lines[line_no - 1].strip()

    assert actual_text == evidence["text"].strip(), (
        f"Source evidence mismatch: {source_id}:{line_no}"
    )

    assert evidence["source_blob_sha"] == (
        P2AUD_ROADMAP_INDEX[source_id]["git_blob_sha"]
    ), f"Source SHA mismatch: {source_id}"

    first = max(1, line_no - context_radius)
    last = min(len(source_lines), line_no + context_radius)

    context = [
        {
            "line": current_line,
            "text": source_lines[current_line - 1].strip()[:250],
        }
        for current_line in range(first, last + 1)
    ]

    return {
        "source_id": source_id,
        "source_path": evidence["source_path"],
        "source_blob_sha": evidence["source_blob_sha"],
        "line": line_no,
        "verified_text": actual_text,
        "context": context,
        "verification_status": "EXACT_LINE_MATCH",
    }


def p2aud_c007_find_historical_matches(
    review_id,
    max_total=10,
    max_per_file=2,
):
    patterns = P2AUD_C007_SEARCH_PATTERNS[review_id]
    compiled = [
        re.compile(pattern, re.IGNORECASE)
        for pattern in patterns
    ]

    matches = []

    for relative_path, content in (
        P2AUD_C007_HISTORICAL_TEXTS.items()
    ):
        local_matches = 0

        for line_no, line in enumerate(
            content.splitlines(),
            start=1,
        ):
            if not any(regex.search(line) for regex in compiled):
                continue

            matches.append({
                "source_path": relative_path,
                "source_sha256": next(
                    item["sha256"]
                    for item in P2AUD_C007_HISTORICAL_INVENTORY
                    if item["relative_path"] == relative_path
                ),
                "line": line_no,
                "text_preview": line.strip()[:240],
                "evidence_status": (
                    "SEARCH_HIT_NOT_APPROVED_DECISION"
                ),
            })

            local_matches += 1

            if (
                local_matches >= max_per_file
                or len(matches) >= max_total
            ):
                break

        if len(matches) >= max_total:
            break

    return matches


P2AUD_C007_REVIEW_DOCKETS = []

for review_item in P2AUD_C006_REVIEW_ITEMS:
    review_id = review_item["review_id"]

    assert review_id in P2AUD_C007_SEARCH_PATTERNS
    assert review_id in P2AUD_C007_REVIEW_HINTS

    primary_evidence = [
        p2aud_c007_primary_evidence(item)
        for item in review_item["evidence"]
    ]

    historical_matches = (
        p2aud_c007_find_historical_matches(review_id)
    )

    docket = {
        "review_id": review_id,
        "topic": review_item["topic"],
        "review_type": review_item["review_type"],
        "description": review_item["description"],
        "primary_evidence": primary_evidence,
        "primary_evidence_verified": len(primary_evidence),
        "historical_matches": historical_matches,
        "historical_match_count": len(historical_matches),
        "initial_review_hint": (
            P2AUD_C007_REVIEW_HINTS[review_id]
        ),
        "initial_review_hint_status": (
            "ANALYST_HYPOTHESIS_NOT_DECISION"
        ),
        "approved_decision_found": "NOT_VERIFIED",
        "conflict_decision": "NOT_ADJUDICATED",
        "approved_resolution": None,
        "next_action": "CONTENT_AND_DECISION_REVIEW",
    }

    P2AUD_C007_REVIEW_DOCKETS.append(docket)


P2AUD_C007_IDS = [
    item["review_id"]
    for item in P2AUD_C007_REVIEW_DOCKETS
]

assert len(P2AUD_C007_IDS) == 5
assert len(P2AUD_C007_IDS) == len(set(P2AUD_C007_IDS))

assert all(
    item["conflict_decision"] == "NOT_ADJUDICATED"
    for item in P2AUD_C007_REVIEW_DOCKETS
)

P2AUD_C007_SOURCE_STATUS = Counter(
    item["status"]
    for item in P2AUD_C007_HISTORICAL_INVENTORY
)

P2AUD_C007_REPORT = {
    "schema_version": "P2AUD_ROADMAP_DIFFERENCE_REVIEW_V1",
    "cell_id": "P2AUD-C007",
    "action": 7,
    "status": (
        "P2AUD_C007_EVIDENCE_COLLECTED_DECISIONS_PENDING"
    ),
    "review_cases": len(P2AUD_C007_REVIEW_DOCKETS),
    "primary_evidence_lines_verified": sum(
        item["primary_evidence_verified"]
        for item in P2AUD_C007_REVIEW_DOCKETS
    ),
    "historical_files_inventory": (
        P2AUD_C007_HISTORICAL_INVENTORY
    ),
    "historical_file_status_counts": dict(
        sorted(P2AUD_C007_SOURCE_STATUS.items())
    ),
    "review_summary": [
        {
            "review_id": item["review_id"],
            "topic": item["topic"],
            "initial_review_hint": item["initial_review_hint"],
            "primary_evidence_verified": (
                item["primary_evidence_verified"]
            ),
            "historical_match_count": (
                item["historical_match_count"]
            ),
            "historical_examples": (
                item["historical_matches"][:3]
            ),
            "decision_status": item["conflict_decision"],
        }
        for item in P2AUD_C007_REVIEW_DOCKETS
    ],
    "requirements_evaluated": 0,
    "conflicts_finally_adjudicated": 0,
    "formal_gates_passed": 0,
    "files_written": 0,
    "new_api_calls": 0,
    "git_operations": 0,
    "next_cell": "P2AUD-C008",
}

assert P2AUD_C007_REPORT["review_cases"] == 5
assert P2AUD_C007_REPORT["conflicts_finally_adjudicated"] == 0

print(
    json.dumps(
        P2AUD_C007_REPORT,
        ensure_ascii=False,
        indent=2,
    )
)
````

````python
# P2AUD-C008 — Build evidence-backed content and decision dossiers
# Action 8 | READ_ONLY | NO_API | NO_FILE_WRITES | NO_GIT_WRITES

assert P2AUD_C007_REPORT["status"] == (
    "P2AUD_C007_EVIDENCE_COLLECTED_DECISIONS_PENDING"
)
assert P2AUD_C007_REPORT["review_cases"] == 5
assert P2AUD_C007_REPORT["primary_evidence_lines_verified"] == 13
assert P2AUD_C007_REPORT["conflicts_finally_adjudicated"] == 0

# Verify the provisional crosswalk has not changed since C006.
P2AUD_C008_CURRENT_CROSSWALK_HASH = hashlib.sha256(
    json.dumps(
        P2AUD_C006_CROSSWALK,
        ensure_ascii=False,
        sort_keys=True,
        separators=(",", ":"),
    ).encode("utf-8")
).hexdigest()

assert P2AUD_C008_CURRENT_CROSSWALK_HASH == (
    P2AUD_C006_REPORT["crosswalk_sha256"]
)

# Interpretations below are proposals, not approved decisions.
P2AUD_C008_CASE_SPECS = {
    "P2AUD-RV-001": {
        "difference_type": "POTENTIAL_THRESHOLD_CONFLICT",
        "focus_units": [
            ("PHASE2_MODULAR", "NOTEBOOK", "NB19"),
            ("PHASE2_OPERATIONAL", "PACKAGE", "P2-11"),
        ],
        "search_terms": (
            r"citation|استناد|claim|پوشش|coverage"
        ),
        "working_interpretation": (
            "The two roadmaps specify different initial "
            "citation coverage targets: 90% and 95%. "
            "Metric denominator and target freeze are unverified."
        ),
        "proposed_resolution": (
            "Define one metric formula, denominator and "
            "evaluation set. Locate approved target-freeze "
            "evidence before selecting an acceptance threshold."
        ),
        "decision_question": (
            "Which citation coverage definition and threshold "
            "were formally approved, and where is the evidence?"
        ),
    },
    "P2AUD-RV-002": {
        "difference_type": "POTENTIAL_SEQUENCE_CONFLICT",
        "focus_units": [
            ("PHASE2_OPERATIONAL", "PACKAGE", "P2-02"),
            ("PHASE2_MODULAR", "NOTEBOOK", "NB01"),
            ("PHASE2_MODULAR", "NOTEBOOK", "NB07"),
        ],
        "search_terms": (
            r"schema|contract|lineage|stable.?id|"
            r"قرارداد|شِما|ورودی|خروجی|نسخه"
        ),
        "working_interpretation": (
            "The operational plan places data contracts "
            "before source pipelines, while the modular "
            "flow places the unified schema notebook after "
            "source ingestion."
        ),
        "proposed_resolution": (
            "Check whether a preliminary source contract "
            "existed before ingestion and NB07 only unified "
            "or hardened it. Otherwise register the ordering "
            "difference as an unresolved planning conflict."
        ),
        "decision_question": (
            "Was there a versioned prerequisite contract "
            "before NB01-NB06, with verifiable evidence?"
        ),
    },
    "P2AUD-RV-003": {
        "difference_type": "PARALLEL_VS_SEQUENTIAL_PLAN",
        "focus_units": [
            ("PHASE2_OPERATIONAL", "PACKAGE", "P2-07"),
            ("PHASE2_OPERATIONAL", "PACKAGE", "P2-08"),
            ("PHASE2_MODULAR", "NOTEBOOK", "NB11"),
            ("PHASE2_MODULAR", "NOTEBOOK", "NB12"),
        ],
        "search_terms": (
            r"golden|chunk|benchmark|dev|test|"
            r"آزمایش|تست|نمونه|بازبین|قطعه"
        ),
        "working_interpretation": (
            "The operational roadmap allows Golden work "
            "in parallel with chunking; the modular diagram "
            "presents a sequential notebook flow."
        ),
        "proposed_resolution": (
            "Distinguish parallel preparation and annotation "
            "from final versioned outputs and acceptance "
            "dependencies. Preserve both where compatible."
        ),
        "decision_question": (
            "Which Golden activities may run in parallel, "
            "and which must wait for chunking artifacts?"
        ),
    },
    "P2AUD-RV-004": {
        "difference_type": "FUNCTIONAL_OWNERSHIP_REVIEW",
        "focus_units": [
            ("PHASE2_MODULAR", "NOTEBOOK", "NB10"),
            ("PHASE2_OPERATIONAL", "PACKAGE", "P2-14"),
        ],
        "search_terms": (
            r"taxonomy|ontology|competency|knowledge|"
            r"دانش|موضوع|مفهوم|مستند|evidence"
        ),
        "working_interpretation": (
            "NB10 is a dedicated taxonomy and knowledge "
            "engineering notebook, without a dedicated "
            "primary operational package in the proposed map."
        ),
        "proposed_resolution": (
            "Record NB10 as a cross-cutting or separately "
            "owned functional deliverable if the roadmap "
            "supports that reading. Do not equate P2-14 "
            "documentation with functional acceptance."
        ),
        "decision_question": (
            "Which operational requirement owns NB10 "
            "taxonomy deliverables and acceptance evidence?"
        ),
    },
    "P2AUD-RV-005": {
        "difference_type": "GLOBAL_GATE_VS_PACKAGE_SCOPE",
        "focus_units": [
            ("PHASE2_MODULAR", "NOTEBOOK", "NB23"),
            ("PHASE2_OPERATIONAL", "PACKAGE", "P2-13"),
        ],
        "search_terms": (
            r"handoff|contract|readiness|release|"
            r"تحویل|فاز.?۳|فاز.?3|قرارداد|آمادگی|gate"
        ),
        "working_interpretation": (
            "NB23 has no dedicated operational package, "
            "but the operational roadmap contains a "
            "separate final Phase 3 handoff gate."
        ),
        "proposed_resolution": (
            "Cross-reference NB23 directly to the global "
            "Phase 3 handoff gate, without adding an "
            "unapproved fifteenth operational package."
        ),
        "decision_question": (
            "Does the global handoff gate contain the full "
            "NB23 contract and readiness requirements?"
        ),
    },
}

P2AUD_C008_EXPECTED_IDS = {
    f"P2AUD-RV-{number:03d}"
    for number in range(1, 6)
}

assert set(P2AUD_C008_CASE_SPECS) == P2AUD_C008_EXPECTED_IDS

P2AUD_C008_UNIT_SOURCES = {
    "NOTEBOOK": P2AUD_C003_UNITS["notebooks"],
    "PACKAGE": P2AUD_C003_UNITS["operational_packages"],
}

P2AUD_C008_SCOPE_PATTERN = re.compile(
    r"target|baseline|freeze|فریز|"
    r"پیشنهاد|اولیه|تأیید|تصویب",
    re.IGNORECASE,
)


def p2aud_c008_extract_focus_unit(
    source_id,
    unit_type,
    unit_id,
    search_terms,
):
    unit = P2AUD_C008_UNIT_SOURCES[
        unit_type
    ][unit_id]

    source_lines = P2AUD_ROADMAP_TEXTS[
        source_id
    ].splitlines()

    start = unit["start_line"]
    end = unit["end_line"]
    regex = re.compile(
        search_terms,
        re.IGNORECASE,
    )

    hits = []

    for number in range(start, end + 1):
        text = source_lines[number - 1].strip()

        if not text:
            continue

        if regex.search(text):
            hits.append({
                "line": number,
                "text": text[:250],
            })

    return {
        "source_id": source_id,
        "source_blob_sha": (
            P2AUD_ROADMAP_INDEX[source_id][
                "git_blob_sha"
            ]
        ),
        "unit_type": unit_type,
        "unit_id": unit_id,
        "unit_title": unit["title"],
        "start_line": start,
        "end_line": end,
        "section_sha256": unit["section_sha256"],
        "keyword_hit_count": len(hits),
        "keyword_examples": hits[:6],
        "review_status": "SUPPORTING_CONTEXT_ONLY",
    }


def p2aud_c008_expand_primary_evidence(item):
    source_id = item["source_id"]
    line_no = item["line"]

    lines = P2AUD_ROADMAP_TEXTS[
        source_id
    ].splitlines()

    assert lines[line_no - 1].strip() == (
        item["verified_text"]
    )

    assert item["verification_status"] == (
        "EXACT_LINE_MATCH"
    )

    first = max(1, line_no - 20)
    last = min(len(lines), line_no + 5)

    planning_notes = [
        {
            "line": number,
            "text": lines[number - 1].strip()[:220],
        }
        for number in range(first, last + 1)
        if (
            number != line_no
            and P2AUD_C008_SCOPE_PATTERN.search(
                lines[number - 1]
            )
        )
    ]

    return {
        "source_id": source_id,
        "source_path": item["source_path"],
        "source_blob_sha": item["source_blob_sha"],
        "line": line_no,
        "text": item["verified_text"],
        "context": item["context"],
        "nearby_planning_notes": planning_notes[:5],
        "verification": "SOURCE_LINE_VERIFIED",
    }


P2AUD_C008_CONTENT_DOCKETS = []
P2AUD_C008_DECISION_REGISTER = []

for previous_docket in P2AUD_C007_REVIEW_DOCKETS:
    review_id = previous_docket["review_id"]
    spec = P2AUD_C008_CASE_SPECS[review_id]

    primary_evidence = [
        p2aud_c008_expand_primary_evidence(item)
        for item in previous_docket["primary_evidence"]
    ]

    focus_sections = [
        p2aud_c008_extract_focus_unit(
            source_id,
            unit_type,
            unit_id,
            spec["search_terms"],
        )
        for source_id, unit_type, unit_id
        in spec["focus_units"]
    ]

    historical_examples = [
        {
            "source_path": item["source_path"],
            "source_sha256": item["source_sha256"],
            "line": item["line"],
            "text_preview": item["text_preview"],
            "status": (
                "HISTORICAL_SEARCH_HIT_NOT_DECISION"
            ),
        }
        for item in previous_docket[
            "historical_matches"
        ][:4]
    ]

    evidence_identity = {
        "review_id": review_id,
        "primary": [
            (
                item["source_id"],
                item["source_blob_sha"],
                item["line"],
            )
            for item in primary_evidence
        ],
        "focus": [
            (
                item["source_id"],
                item["unit_id"],
                item["section_sha256"],
            )
            for item in focus_sections
        ],
    }

    evidence_sha256 = hashlib.sha256(
        json.dumps(
            evidence_identity,
            ensure_ascii=False,
            sort_keys=True,
        ).encode("utf-8")
    ).hexdigest()

    docket = {
        "review_id": review_id,
        "topic": previous_docket["topic"],
        "difference_type": spec["difference_type"],
        "primary_evidence": primary_evidence,
        "supporting_sections": focus_sections,
        "historical_examples": historical_examples,
        "total_historical_search_hits": (
            previous_docket["historical_match_count"]
        ),
        "working_interpretation": (
            spec["working_interpretation"]
        ),
        "proposed_resolution": (
            spec["proposed_resolution"]
        ),
        "decision_question": (
            spec["decision_question"]
        ),
        "evidence_sha256": evidence_sha256,
        "content_review_status": (
            "EVIDENCE_STAGED_FOR_MANUAL_REVIEW"
        ),
        "final_classification": "NOT_VERIFIED",
        "approved_resolution": None,
    }

    P2AUD_C008_CONTENT_DOCKETS.append(docket)

    P2AUD_C008_DECISION_REGISTER.append({
        "review_id": review_id,
        "topic": previous_docket["topic"],
        "difference_type": spec["difference_type"],
        "decision_question": spec["decision_question"],
        "proposed_resolution": spec["proposed_resolution"],
        "classification_options": [
            "CONSISTENT",
            "DIFFERENT_SCOPE",
            "CONFLICT",
            "SUPERSEDED_WITH_EVIDENCE",
            "NOT_VERIFIED",
        ],
        "final_classification": "NOT_VERIFIED",
        "approved_resolution": None,
        "approval_status": "PENDING_EVIDENCE_REVIEW",
        "evidence_sha256": evidence_sha256,
    })


assert len(P2AUD_C008_CONTENT_DOCKETS) == 5
assert len(P2AUD_C008_DECISION_REGISTER) == 5

assert sum(
    len(item["primary_evidence"])
    for item in P2AUD_C008_CONTENT_DOCKETS
) == 13

assert all(
    item["final_classification"] == "NOT_VERIFIED"
    and item["approved_resolution"] is None
    for item in P2AUD_C008_DECISION_REGISTER
)

P2AUD_C008_TYPE_COUNTS = Counter(
    item["difference_type"]
    for item in P2AUD_C008_DECISION_REGISTER
)

P2AUD_C008_REPORT = {
    "schema_version": (
        "P2AUD_CONTENT_RECONCILIATION_DOCKETS_V1"
    ),
    "cell_id": "P2AUD-C008",
    "action": 8,
    "status": (
        "P2AUD_C008_CONTENT_DOCKETS_READY_DECISIONS_OPEN"
    ),
    "review_cases": len(
        P2AUD_C008_CONTENT_DOCKETS
    ),
    "primary_evidence_verified": 13,
    "supporting_sections_examined": sum(
        len(item["supporting_sections"])
        for item in P2AUD_C008_CONTENT_DOCKETS
    ),
    "difference_type_counts": dict(sorted(
        P2AUD_C008_TYPE_COUNTS.items()
    )),
    "review_summary": [
        {
            "review_id": item["review_id"],
            "topic": item["topic"],
            "difference_type": item["difference_type"],
            "working_interpretation": (
                item["working_interpretation"]
            ),
            "proposed_resolution": (
                item["proposed_resolution"]
            ),
            "decision_question": (
                item["decision_question"]
            ),
            "primary_evidence_count": len(
                item["primary_evidence"]
            ),
            "supporting_sections": [
                {
                    "unit": section["unit_id"],
                    "source": section["source_id"],
                    "keyword_hits": (
                        section["keyword_hit_count"]
                    ),
                    "examples": (
                        section["keyword_examples"][:2]
                    ),
                }
                for section in item["supporting_sections"]
            ],
            "historical_search_hits": (
                item["total_historical_search_hits"]
            ),
            "final_classification": (
                item["final_classification"]
            ),
        }
        for item in P2AUD_C008_CONTENT_DOCKETS
    ],
    "final_decisions_approved": 0,
    "formal_gates_passed": 0,
    "requirements_evaluated": 0,
    "files_written": 0,
    "new_api_calls": 0,
    "git_operations": 0,
    "next_cell": "P2AUD-C009",
}

print(
    json.dumps(
        P2AUD_C008_REPORT,
        ensure_ascii=False,
        indent=2,
    )
)
````

````python
# P2AUD-C009 — Evidence-backed provisional roadmap difference triage
# Action 9 | READ_ONLY | NO_API | NO_FILE_WRITES | NO_GIT_WRITES

assert P2AUD_C008_REPORT["status"] == (
    "P2AUD_C008_CONTENT_DOCKETS_READY_DECISIONS_OPEN"
)
assert P2AUD_C008_REPORT["review_cases"] == 5
assert P2AUD_C008_REPORT["final_decisions_approved"] == 0

P2AUD_C009_CASE_POLICY = {
    "P2AUD-RV-001": {
        "hypothesis": "POTENTIAL_THRESHOLD_CONFLICT",
        "proposed_classification": "CONFLICT",
        "reason": (
            "Different numerical coverage targets appear in "
            "the two roadmaps. Their denominator, applicability "
            "and approved freeze status remain unverified."
        ),
        "required_evidence": (
            "Approved metric definition, denominator, "
            "evaluation set and frozen threshold."
        ),
        "proposed_action": (
            "Reconcile the metric definition first; do not "
            "silently select either 90% or 95%."
        ),
    },
    "P2AUD-RV-002": {
        "hypothesis": "DEPENDENCY_ORDER_REVIEW",
        "proposed_classification": "NOT_VERIFIED",
        "reason": (
            "Operational contracts precede ingestion, "
            "but the unified schema notebook follows "
            "source ingestion in the modular flow."
        ),
        "required_evidence": (
            "Versioned prerequisite contract and its "
            "actual use by NB01-NB06."
        ),
        "proposed_action": (
            "Separate prerequisite source contracts "
            "from later schema consolidation if supported."
        ),
    },
    "P2AUD-RV-003": {
        "hypothesis": "POSSIBLY_COMPATIBLE_EXECUTION_SCOPES",
        "proposed_classification": "DIFFERENT_SCOPE",
        "reason": (
            "Parallel Golden preparation may coexist "
            "with a sequential artifact acceptance flow. "
            "Actual dependencies have not been verified."
        ),
        "required_evidence": (
            "Golden annotation prerequisites, chunk "
            "dependencies and versioned release contracts."
        ),
        "proposed_action": (
            "Document which Golden activities can run "
            "in parallel and which require chunk outputs."
        ),
    },
    "P2AUD-RV-004": {
        "hypothesis": "POSSIBLE_FUNCTIONAL_OWNERSHIP_GAP",
        "proposed_classification": "NOT_VERIFIED",
        "reason": (
            "NB10 has no primary operational package "
            "in the proposed crosswalk. Documentation "
            "ownership alone does not prove functional coverage."
        ),
        "required_evidence": (
            "Explicit taxonomy functional owner, "
            "deliverables and acceptance evidence."
        ),
        "proposed_action": (
            "Register NB10 as an explicit cross-cutting "
            "functional responsibility if supported."
        ),
    },
    "P2AUD-RV-005": {
        "hypothesis": "POSSIBLE_GLOBAL_GATE_ALIGNMENT",
        "proposed_classification": "DIFFERENT_SCOPE",
        "reason": (
            "The operational roadmap contains a "
            "global Phase 3 handoff gate, while NB23 "
            "defines its detailed notebook implementation."
        ),
        "required_evidence": (
            "Complete comparison of the NB23 contracts "
            "with the operational final handoff gate."
        ),
        "proposed_action": (
            "Connect NB23 to the existing global gate "
            "without inventing a fifteenth package."
        ),
    },
}

P2AUD_C009_EXPECTED_IDS = {
    f"P2AUD-RV-{number:03d}"
    for number in range(1, 6)
}

assert set(P2AUD_C009_CASE_POLICY) == P2AUD_C009_EXPECTED_IDS

P2AUD_C009_VALID_CLASSIFICATIONS = {
    "CONSISTENT",
    "DIFFERENT_SCOPE",
    "CONFLICT",
    "SUPERSEDED_WITH_EVIDENCE",
    "NOT_VERIFIED",
}

P2AUD_C009_DIGIT_TRANSLATION = str.maketrans(
    "۰۱۲۳۴۵۶۷۸۹٠١٢٣٤٥٦٧٨٩",
    "01234567890123456789",
)


def p2aud_c009_numeric_percent(text):
    normalized = (
        text.translate(P2AUD_C009_DIGIT_TRANSLATION)
        .replace("٪", "%")
        .replace("٫", ".")
    )

    decimal_match = re.search(
        r"(?<!\d)0[.,](\d+)(?!\d)",
        normalized,
    )

    if decimal_match:
        number = float(
            "0." + decimal_match.group(1)
        )
        return round(number * 100, 6)

    percent_match = re.search(
        r"(?<!\d)(\d+(?:[.,]\d+)?)\s*%",
        normalized,
    )

    if percent_match:
        return float(
            percent_match.group(1).replace(",", ".")
        )

    return None


P2AUD_C009_EVIDENCE_LEDGER = []
P2AUD_C009_TRIAGE_REGISTER = []

for docket in P2AUD_C008_CONTENT_DOCKETS:
    review_id = docket["review_id"]
    policy = P2AUD_C009_CASE_POLICY[review_id]

    verified_evidence = []
    all_lines_valid = True

    for evidence in docket["primary_evidence"]:
        source_id = evidence["source_id"]
        source_line = evidence["line"]

        actual_lines = P2AUD_ROADMAP_TEXTS[
            source_id
        ].splitlines()

        actual_text = actual_lines[
            source_line - 1
        ].strip()

        expected_sha = P2AUD_ROADMAP_INDEX[
            source_id
        ]["git_blob_sha"]

        line_valid = (
            actual_text == evidence["text"]
            and expected_sha == evidence["source_blob_sha"]
        )

        all_lines_valid = (
            all_lines_valid and line_valid
        )

        candidate_matches = [
            candidate["candidate_id"]
            for candidate in P2AUD_C004_CANDIDATES
            if (
                candidate["source_id"] == source_id
                and candidate["source_line"] == source_line
            )
        ]

        verified_evidence.append({
            "source_id": source_id,
            "source_path": evidence["source_path"],
            "source_blob_sha": expected_sha,
            "line": source_line,
            "text": actual_text,
            "source_verification": (
                "PASS" if line_valid else "FAIL"
            ),
            "candidate_ids": candidate_matches,
            "candidate_index_status": (
                "INDEXED_IN_C004"
                if candidate_matches
                else "DIRECT_SOURCE_EVIDENCE_ONLY"
            ),
        })

    if not all_lines_valid:
        raise AssertionError(
            f"Primary evidence changed: {review_id}"
        )

    P2AUD_C009_EVIDENCE_LEDGER.extend(
        {
            "review_id": review_id,
            **item,
        }
        for item in verified_evidence
    )

    observations = {}

    if review_id == "P2AUD-RV-001":
        thresholds = {}

        for evidence in verified_evidence:
            value = p2aud_c009_numeric_percent(
                evidence["text"]
            )
            if value is not None:
                thresholds[
                    evidence["source_id"]
                ] = value

        if set(thresholds) != {
            "PHASE2_MODULAR",
            "PHASE2_OPERATIONAL",
        }:
            raise ValueError(
                "Could not verify both citation thresholds"
            )

        observations = {
            "modular_target_percent": thresholds[
                "PHASE2_MODULAR"
            ],
            "operational_target_percent": thresholds[
                "PHASE2_OPERATIONAL"
            ],
            "difference_percentage_points": round(
                abs(
                    thresholds["PHASE2_MODULAR"]
                    - thresholds["PHASE2_OPERATIONAL"]
                ),
                6,
            ),
            "metric_definition_equivalence": (
                "NOT_VERIFIED"
            ),
            "target_freeze_evidence": "NOT_VERIFIED",
        }

    elif review_id == "P2AUD-RV-002":
        operational_lines = [
            item["line"]
            for item in verified_evidence
            if item["source_id"] == "PHASE2_OPERATIONAL"
        ]

        modular_lines = [
            item["line"]
            for item in verified_evidence
            if item["source_id"] == "PHASE2_MODULAR"
        ]

        observations = {
            "operational_sequence_lines": (
                operational_lines
            ),
            "modular_sequence_lines": modular_lines,
            "contract_before_ingestion_in_operational": (
                len(operational_lines) >= 2
                and min(operational_lines)
                < max(operational_lines)
            ),
            "unified_schema_after_ingestion_in_modular": (
                len(modular_lines) >= 2
                and min(modular_lines)
                < max(modular_lines)
            ),
            "prerequisite_contract_verified": False,
        }

    elif review_id == "P2AUD-RV-003":
        operational_text = " ".join(
            item["text"]
            for item in verified_evidence
            if item["source_id"] == "PHASE2_OPERATIONAL"
        )

        modular_text = " ".join(
            item["text"]
            for item in verified_evidence
            if item["source_id"] == "PHASE2_MODULAR"
        )

        observations = {
            "operational_parallel_word_detected": (
                "هم‌زمان" in operational_text
            ),
            "modular_sequence_arrows_detected": (
                "-->" in modular_text
            ),
            "artifact_dependency_compatibility": (
                "NOT_VERIFIED"
            ),
        }

    elif review_id == "P2AUD-RV-004":
        nb10_primary_packages = [
            row["package_id"]
            for row in P2AUD_C006_CROSSWALK
            if "NB10" in row["primary_notebooks"]
        ]

        nb10_support_packages = [
            row["package_id"]
            for row in P2AUD_C006_CROSSWALK
            if "NB10" in row["support_notebooks"]
        ]

        observations = {
            "nb10_primary_packages": nb10_primary_packages,
            "nb10_support_packages": nb10_support_packages,
            "functional_ownership_accepted": False,
            "absence_is_implementation_failure": False,
        }

    elif review_id == "P2AUD-RV-005":
        nb23_primary_packages = [
            row["package_id"]
            for row in P2AUD_C006_CROSSWALK
            if "NB23" in row["primary_notebooks"]
        ]

        operational_global_sections = (
            P2AUD_C004_GLOBAL_SECTIONS[
                "PHASE2_OPERATIONAL"
            ]
        )

        handoff_sections = [
            {
                "unit_id": unit_id,
                "start_line": unit["start_line"],
                "end_line": unit["end_line"],
                "title": unit["title"],
            }
            for unit_id, unit
            in operational_global_sections.items()
            if (
                "Gate نهایی" in unit["title"]
                and "فاز ۳" in unit["title"]
            )
        ]

        observations = {
            "nb23_primary_packages": nb23_primary_packages,
            "global_handoff_sections": handoff_sections,
            "global_handoff_section_found": bool(
                handoff_sections
            ),
            "complete_contract_alignment": "NOT_VERIFIED",
        }

    proposed = policy["proposed_classification"]

    assert proposed in (
        P2AUD_C009_VALID_CLASSIFICATIONS
    )

    triage_record = {
        "review_id": review_id,
        "topic": docket["topic"],
        "hypothesis": policy["hypothesis"],
        "proposed_classification": proposed,
        "proposal_status": (
            "ANALYST_PROPOSAL_NOT_APPROVED"
        ),
        "reason": policy["reason"],
        "observations": observations,
        "required_evidence": policy["required_evidence"],
        "proposed_action": policy["proposed_action"],
        "decision_question": docket["decision_question"],
        "primary_evidence_count": len(
            verified_evidence
        ),
        "primary_evidence_verified": all_lines_valid,
        "historical_search_hits": (
            docket["total_historical_search_hits"]
        ),
        "historical_decision_approved": "NOT_VERIFIED",
        "final_classification": "NOT_VERIFIED",
        "approved_resolution": None,
        "approval_status": "PENDING",
    }

    P2AUD_C009_TRIAGE_REGISTER.append(
        triage_record
    )


assert len(P2AUD_C009_TRIAGE_REGISTER) == 5
assert len(P2AUD_C009_EVIDENCE_LEDGER) == 13

assert all(
    record["primary_evidence_verified"]
    for record in P2AUD_C009_TRIAGE_REGISTER
)

assert all(
    record["final_classification"] == "NOT_VERIFIED"
    and record["approved_resolution"] is None
    and record["approval_status"] == "PENDING"
    for record in P2AUD_C009_TRIAGE_REGISTER
)

P2AUD_C009_PROPOSAL_COUNTS = Counter(
    record["proposed_classification"]
    for record in P2AUD_C009_TRIAGE_REGISTER
)

P2AUD_C009_EVIDENCE_STATUS_COUNTS = Counter(
    record["candidate_index_status"]
    for record in P2AUD_C009_EVIDENCE_LEDGER
)

P2AUD_C009_REPORT = {
    "schema_version": (
        "P2AUD_PROVISIONAL_DIFFERENCE_TRIAGE_V1"
    ),
    "cell_id": "P2AUD-C009",
    "action": 9,
    "status": (
        "P2AUD_C009_PROVISIONAL_TRIAGE_READY"
        "_DECISIONS_PENDING"
    ),
    "review_cases": len(P2AUD_C009_TRIAGE_REGISTER),
    "primary_evidence_lines_reverified": len(
        P2AUD_C009_EVIDENCE_LEDGER
    ),
    "evidence_index_status_counts": dict(
        sorted(P2AUD_C009_EVIDENCE_STATUS_COUNTS.items())
    ),
    "proposal_classification_counts": dict(
        sorted(P2AUD_C009_PROPOSAL_COUNTS.items())
    ),
    "triage_summary": [
        {
            "review_id": record["review_id"],
            "topic": record["topic"],
            "proposed_classification": (
                record["proposed_classification"]
            ),
            "proposal_status": record["proposal_status"],
            "observations": record["observations"],
            "required_evidence": record["required_evidence"],
            "approved_resolution": (
                record["approved_resolution"]
            ),
            "final_classification": (
                record["final_classification"]
            ),
        }
        for record in P2AUD_C009_TRIAGE_REGISTER
    ],
    "provisional_cases_reviewed": 5,
    "final_decisions_approved": 0,
    "final_conflicts_adjudicated": 0,
    "formal_gates_passed": 0,
    "files_written": 0,
    "new_api_calls": 0,
    "git_operations": 0,
    "next_cell": "P2AUD-C010",
}

print(
    json.dumps(
        P2AUD_C009_REPORT,
        ensure_ascii=False,
        indent=2,
    )
)
````

- نبود عبارت در محدودهٔ جست‌وجو، به معنای نبود الزام در کل پروژه نیست.
- وجود متن الزام، اثبات انجام‌شدن آن نیست.
- تصمیم نهایی هر پنج پرونده همچنان باز می‌ماند.

### خروجی‌ها

- `P2AUD_C010_SCOPES`
- `P2AUD_C010_REQUIREMENT_CHECKS`
- `P2AUD_C010_CASE_DOSSIERS`
- `P2AUD_C010_REPORT`

### شرط پایان

پنج پرونده با کنترل‌های مستند و شروط بررسی تکمیلی آماده شوند. هیچ Gate یا تصمیمی بدون شاهد اجرایی و تأیید لازم PASS نمی‌شود.

**محدودیت اجرا:** فقط خواندن داده‌های موجود در حافظه؛ بدون API، تغییر Git یا ایجاد فایل.

</div>

````python
# P2AUD-C010 — Build traceable requirement and decision-readiness matrix
# Action 10 | READ_ONLY | NO_API | NO_FILE_WRITES | NO_GIT_WRITES

assert P2AUD_C009_REPORT["status"] == (
    "P2AUD_C009_PROVISIONAL_TRIAGE_READY_DECISIONS_PENDING"
)
assert P2AUD_C009_REPORT["review_cases"] == 5
assert P2AUD_C009_REPORT["final_decisions_approved"] == 0

P2AUD_C010_CASE_SPECS = {
    "P2AUD-RV-001": {
        "scopes": [
            ("PHASE2_MODULAR", "NOTEBOOK", "NB19"),
            ("PHASE2_OPERATIONAL", "PACKAGE", "P2-11"),
            ("PHASE2_MODULAR", "GLOBAL", "معیارهای اولیه پذیرش"),
            ("PHASE2_OPERATIONAL", "GLOBAL", "معیارهای پیشنهادی اولیه"),
        ],
        "checks": [
            {
                "id": "CITATION_TARGETS",
                "label": "آستانه‌های پوشش Citation",
                "pattern": r"citation.{0,45}(coverage|correctness)|پوشش.{0,30}استناد",
            },
            {
                "id": "CITATION_VERIFICATION",
                "label": "سنجش و صحت Citation",
                "pattern": r"citation|claim|استناد|اعتبارسنجی",
            },
            {
                "id": "TARGET_FREEZE",
                "label": "شرط نهایی‌سازی معیار",
                "pattern": r"target|baseline|freeze|فریز|اولیه|تأیید",
            },
        ],
        "implementation_evidence_required": [
            "تعریف دقیق فرمول Citation Coverage",
            "مخرج و دامنهٔ Claimهای مشمول معیار",
            "سند تأیید و Freeze آستانه",
            "نتیجهٔ آزمون روی مجموعهٔ ارزیابی معتبر",
        ],
    },
    "P2AUD-RV-002": {
        "scopes": [
            ("PHASE2_OPERATIONAL", "PACKAGE", "P2-02"),
            ("PHASE2_MODULAR", "NOTEBOOK", "NB01"),
            ("PHASE2_MODULAR", "NOTEBOOK", "NB07"),
            ("PHASE2_OPERATIONAL", "GLOBAL", "ترتیب اجرای واقعی"),
            ("PHASE2_MODULAR", "GLOBAL", "فلوچارت کل اجرای"),
        ],
        "checks": [
            {
                "id": "DATA_CONTRACT",
                "label": "قرارداد داده و Schema",
                "pattern": r"contract|schema|قرارداد|stable.?id|lineage",
            },
            {
                "id": "DEPENDENCY_ORDER",
                "label": "ترتیب اجرای قرارداد و Ingestion",
                "pattern": r"P2-02|P2-03|NB07|F\s*-->\s*G|G\s*-->\s*H",
            },
            {
                "id": "SOURCE_VALIDATION",
                "label": "اعتبارسنجی ورودی و نسخه",
                "pattern": r"version|hash|schema|ورودی|نسخه|شناسه",
            },
        ],
        "implementation_evidence_required": [
            "قرارداد نسخه‌دار موجود پیش از Ingestion",
            "شواهد استفادهٔ NB01 تا NB06 از قرارداد اولیه",
            "نقش واقعی NB07 در یکپارچه‌سازی قراردادها",
            "تست‌های Schema و Migration",
        ],
    },
    "P2AUD-RV-003": {
        "scopes": [
            ("PHASE2_OPERATIONAL", "PACKAGE", "P2-07"),
            ("PHASE2_OPERATIONAL", "PACKAGE", "P2-08"),
            ("PHASE2_MODULAR", "NOTEBOOK", "NB11"),
            ("PHASE2_MODULAR", "NOTEBOOK", "NB12"),
            ("PHASE2_OPERATIONAL", "GLOBAL", "ترتیب اجرای واقعی"),
            ("PHASE2_MODULAR", "GLOBAL", "فلوچارت کل اجرای"),
        ],
        "checks": [
            {
                "id": "GOLDEN_REQUIREMENTS",
                "label": "الزامات Golden Dataset",
                "pattern": r"golden|question|سؤال|annotation|بازبین",
            },
            {
                "id": "CHUNKING_DEPENDENCIES",
                "label": "وابستگی به Chunking",
                "pattern": r"chunk|قطعه|locator|benchmark|artifact",
            },
            {
                "id": "EXECUTION_SEQUENCE",
                "label": "اجرای ترتیبی یا موازی",
                "pattern": r"هم‌زمان|موازی|P2-08|L\s*-->\s*M",
            },
        ],
        "implementation_evidence_required": [
            "تفکیک مراحل آماده‌سازی و Freeze مجموعهٔ Golden",
            "ورودی‌ها و وابستگی‌های واقعی NB12",
            "نسخهٔ Chunk و Locator مرتبط با Gold",
            "مدارک اجرای معتبر و قابل بازتولید",
        ],
    },
    "P2AUD-RV-004": {
        "scopes": [
            ("PHASE2_MODULAR", "NOTEBOOK", "NB10"),
            ("PHASE2_OPERATIONAL", "PACKAGE", "P2-14"),
            ("PHASE2_MODULAR", "GLOBAL", "Evidence Matrix"),
            ("MASTER_ROADMAP", "MASTER", "PHASE2_SCOPE"),
        ],
        "checks": [
            {
                "id": "TAXONOMY_FUNCTION",
                "label": "الزامات کاربردی Taxonomy",
                "pattern": r"taxonomy|competency|topic|ontology|مهندسی دانش",
            },
            {
                "id": "EXPERT_VALIDATION",
                "label": "بازبینی و تأیید متخصص",
                "pattern": r"expert|review|بازبینی|متخصص|تأیید",
            },
            {
                "id": "FUNCTIONAL_DELIVERABLES",
                "label": "خروجی و شواهد مهندسی دانش",
                "pattern": r"coverage|freeze|artifact|دانش|شاهد|evidence",
            },
        ],
        "implementation_evidence_required": [
            "مالک مشخص خروجی‌های کاربردی NB10",
            "Taxonomy و Competency نسخه‌دار",
            "گزارش Coverage و بازبینی متخصص",
            "Gate واقعی و شواهد پذیرش NB10",
        ],
    },
    "P2AUD-RV-005": {
        "scopes": [
            ("PHASE2_MODULAR", "NOTEBOOK", "NB23"),
            ("PHASE2_OPERATIONAL", "PACKAGE", "P2-13"),
            ("PHASE2_OPERATIONAL", "GLOBAL", "Gate نهایی و تحویل"),
            ("PHASE2_MODULAR", "GLOBAL", "تعریف Done کل فاز ۲"),
            ("MASTER_ROADMAP", "MASTER", "PHASE_HANDOFF"),
        ],
        "checks": [
            {
                "id": "HANDOFF_CONTRACT",
                "label": "قرارداد Request و Response",
                "pattern": r"request|response|contract|schema|قرارداد",
            },
            {
                "id": "RELEASE_TRACEABILITY",
                "label": "ردیابی نسخه‌ها و Release",
                "pattern": r"release|version|artifact|dataset|index|نسخه",
            },
            {
                "id": "READINESS_ACCEPTANCE",
                "label": "آمادگی و شرط تحویل",
                "pattern": r"readiness|gate|تست|تحویل|پذیرش|rollback",
            },
        ],
        "implementation_evidence_required": [
            "Request و Response Schema نسخه‌دار",
            "Contract Tests و Backward Compatibility",
            "Release پذیرفته‌شده و نسخه‌های مرتبط",
            "Readiness Report و Gate تحویل واقعی",
        ],
    },
}

P2AUD_C010_EXPECTED_IDS = {
    f"P2AUD-RV-{i:03d}" for i in range(1, 6)
}

assert set(P2AUD_C010_CASE_SPECS) == P2AUD_C010_EXPECTED_IDS

P2AUD_C010_SOURCE_UNITS = {
    "NOTEBOOK": P2AUD_C003_UNITS["notebooks"],
    "PACKAGE": P2AUD_C003_UNITS["operational_packages"],
    "MASTER": P2AUD_C003_UNITS["master_sections"],
}

P2AUD_C010_CANDIDATE_INDEX = defaultdict(list)

for candidate in P2AUD_C004_CANDIDATES:
    key = (
        candidate["source_id"],
        candidate["source_line"],
    )
    P2AUD_C010_CANDIDATE_INDEX[key].append(
        candidate["candidate_id"]
    )


def p2aud_c010_resolve_scope(source_id, unit_type, key):
    if unit_type == "GLOBAL":
        units = P2AUD_C004_GLOBAL_SECTIONS.get(
            source_id, {}
        )

        matches = [
            (unit_id, unit)
            for unit_id, unit in units.items()
            if key.casefold() in unit["title"].casefold()
        ]

        if len(matches) != 1:
            return {
                "status": "SCOPE_NOT_UNIQUELY_FOUND",
                "source_id": source_id,
                "unit_type": unit_type,
                "requested_key": key,
                "match_count": len(matches),
            }

        unit_id, unit = matches[0]

    else:
        units = P2AUD_C010_SOURCE_UNITS[unit_type]
        unit = units.get(key)

        if unit is None:
            return {
                "status": "SCOPE_NOT_FOUND",
                "source_id": source_id,
                "unit_type": unit_type,
                "requested_key": key,
            }

        unit_id = key

    return {
        "status": "FOUND",
        "source_id": source_id,
        "unit_type": unit_type,
        "unit_id": unit_id,
        "title": unit["title"],
        "start_line": unit["start_line"],
        "end_line": unit["end_line"],
        "section_sha256": unit["section_sha256"],
        "source_blob_sha": (
            P2AUD_ROADMAP_INDEX[source_id]["git_blob_sha"]
        ),
    }


P2AUD_C010_SCOPES = {}
P2AUD_C010_REQUIREMENT_CHECKS = []
P2AUD_C010_CASE_DOSSIERS = []

P2AUD_C010_PRIOR_TRIAGE = {
    item["review_id"]: item
    for item in P2AUD_C009_TRIAGE_REGISTER
}

for review_id, spec in P2AUD_C010_CASE_SPECS.items():
    scopes = [
        p2aud_c010_resolve_scope(*scope_spec)
        for scope_spec in spec["scopes"]
    ]

    P2AUD_C010_SCOPES[review_id] = scopes
    case_check_ids = []

    for check_spec in spec["checks"]:
        regex = re.compile(
            check_spec["pattern"],
            flags=re.IGNORECASE,
        )

        matches = []

        for scope in scopes:
            if scope["status"] != "FOUND":
                continue

            source_id = scope["source_id"]
            source_lines = P2AUD_ROADMAP_TEXTS[
                source_id
            ].splitlines()

            for line_no in range(
                scope["start_line"],
                scope["end_line"] + 1,
            ):
                text = source_lines[line_no - 1].strip()

                if not text or not regex.search(text):
                    continue

                candidate_ids = (
                    P2AUD_C010_CANDIDATE_INDEX.get(
                        (source_id, line_no), []
                    )
                )

                matches.append({
                    "source_id": source_id,
                    "source_blob_sha": (
                        scope["source_blob_sha"]
                    ),
                    "unit_id": scope["unit_id"],
                    "line": line_no,
                    "text": text[:300],
                    "candidate_ids": candidate_ids,
                    "source_status": (
                        "INDEXED_IN_C004"
                        if candidate_ids
                        else "DIRECT_SOURCE_ONLY"
                    ),
                })

        check_id = (
            review_id + "-" + check_spec["id"]
        )

        check_record = {
            "check_id": check_id,
            "review_id": review_id,
            "label": check_spec["label"],
            "source_match_count": len(matches),
            "matched_source_ids": sorted({
                item["source_id"]
                for item in matches
            }),
            "evidence_matches": matches,
            "documentation_status": (
                "ROADMAP_TEXT_FOUND"
                if matches
                else "NOT_FOUND_IN_TARGETED_SCOPES"
            ),
            "implementation_status": "NOT_EVALUATED",
            "acceptance_status": "NOT_EVALUATED",
        }

        P2AUD_C010_REQUIREMENT_CHECKS.append(
            check_record
        )
        case_check_ids.append(check_id)

    prior = P2AUD_C010_PRIOR_TRIAGE[review_id]

    P2AUD_C010_CASE_DOSSIERS.append({
        "review_id": review_id,
        "topic": prior["topic"],
        "proposed_classification": (
            prior["proposed_classification"]
        ),
        "scope_count": len(scopes),
        "scopes_found": sum(
            scope["status"] == "FOUND"
            for scope in scopes
        ),
        "check_ids": case_check_ids,
        "implementation_evidence_required": (
            spec["implementation_evidence_required"]
        ),
        "source_interpretation": (
            "PROVISIONAL_NOT_APPROVED"
        ),
        "actual_implementation_review": "PENDING",
        "final_classification": "NOT_VERIFIED",
        "approved_resolution": None,
        "approval_status": "PENDING",
    })


assert len(P2AUD_C010_CASE_DOSSIERS) == 5
assert len(P2AUD_C010_REQUIREMENT_CHECKS) == 15

P2AUD_C010_SCOPE_ERRORS = [
    {
        "review_id": review_id,
        **scope,
    }
    for review_id, scopes in P2AUD_C010_SCOPES.items()
    for scope in scopes
    if scope["status"] != "FOUND"
]

P2AUD_C010_CHECK_STATUS_COUNTS = Counter(
    item["documentation_status"]
    for item in P2AUD_C010_REQUIREMENT_CHECKS
)

P2AUD_C010_DIRECT_SOURCE_COUNT = sum(
    item["source_status"] == "DIRECT_SOURCE_ONLY"
    for check in P2AUD_C010_REQUIREMENT_CHECKS
    for item in check["evidence_matches"]
)

P2AUD_C010_REPORT = {
    "schema_version": (
        "P2AUD_REQUIREMENT_DECISION_READINESS_V1"
    ),
    "cell_id": "P2AUD-C010",
    "action": 10,
    "status": (
        "P2AUD_C010_REQUIREMENT_MATRIX_STAGED"
        if not P2AUD_C010_SCOPE_ERRORS
        else "P2AUD_C010_SCOPE_REVIEW_REQUIRED"
    ),
    "review_cases": len(
        P2AUD_C010_CASE_DOSSIERS
    ),
    "scopes_requested": sum(
        len(scopes)
        for scopes in P2AUD_C010_SCOPES.values()
    ),
    "scopes_found": sum(
        scope["status"] == "FOUND"
        for scopes in P2AUD_C010_SCOPES.values()
        for scope in scopes
    ),
    "scope_errors": P2AUD_C010_SCOPE_ERRORS,
    "requirement_checks": len(
        P2AUD_C010_REQUIREMENT_CHECKS
    ),
    "check_status_counts": dict(
        sorted(P2AUD_C010_CHECK_STATUS_COUNTS.items())
    ),
    "direct_source_only_matches": (
        P2AUD_C010_DIRECT_SOURCE_COUNT
    ),
    "case_summary": [
        {
            "review_id": docket["review_id"],
            "topic": docket["topic"],
            "proposed_classification": (
                docket["proposed_classification"]
            ),
            "scopes_found": docket["scopes_found"],
            "scope_count": docket["scope_count"],
            "checks": [
                {
                    "check_id": check["check_id"],
                    "documentation_status": (
                        check["documentation_status"]
                    ),
                    "source_match_count": (
                        check["source_match_count"]
                    ),
                    "evidence_examples": (
                        check["evidence_matches"][:2]
                    ),
                }
                for check in P2AUD_C010_REQUIREMENT_CHECKS
                if check["review_id"] == docket["review_id"]
            ],
            "implementation_evidence_required": (
                docket["implementation_evidence_required"]
            ),
            "final_classification": (
                docket["final_classification"]
            ),
        }
        for docket in P2AUD_C010_CASE_DOSSIERS
    ],
    "final_decisions_approved": 0,
    "formal_gates_passed": 0,
    "requirements_implementation_evaluated": 0,
    "files_written": 0,
    "new_api_calls": 0,
    "git_operations": 0,
    "next_cell": "P2AUD-C011",
}

assert P2AUD_C010_REPORT["review_cases"] == 5
assert P2AUD_C010_REPORT["final_decisions_approved"] == 0

print(
    json.dumps(
        P2AUD_C010_REPORT,
        ensure_ascii=False,
        indent=2,
    )
)
````

**هزینهٔ API: صفر — عملیات نوشتنی: صفر**

````python
# P2AUD-C011 — Read-only inventory of actual Phase 2 notebooks
# Action 11 | NO_API | NO_FILE_WRITES | NO_GIT_WRITES

assert P2AUD_C010_REPORT["status"] == (
    "P2AUD_C010_REQUIREMENT_MATRIX_STAGED"
)
assert P2AUD_C010_REPORT["scopes_found"] == 24
assert P2AUD_C010_REPORT["requirement_checks"] == 15

P2AUD_C011_ROOT = Path(P2AUD_REPO_ROOT).resolve()
P2AUD_C011_NOTEBOOK_DIR = (
    P2AUD_C011_ROOT / "notebooks" / "phase2"
)

assert P2AUD_C011_NOTEBOOK_DIR.is_dir(), (
    "Phase 2 notebook directory not found"
)

P2AUD_C011_MAX_NOTEBOOK_BYTES = 64 * 1024 * 1024
P2AUD_C011_MAX_FILES = 600

P2AUD_C011_EXPECTED = {}

for notebook_id, unit in sorted(
    P2AUD_C003_UNITS["notebooks"].items()
):
    match = re.search(
        r"`([^`/\\]+\.ipynb)`",
        unit["title"],
        re.IGNORECASE,
    )

    if match is None:
        raise ValueError(
            f"Planned filename not identified: {notebook_id}"
        )

    filename = match.group(1)

    P2AUD_C011_EXPECTED[notebook_id] = {
        "notebook_id": notebook_id,
        "expected_filename": filename,
        "expected_relative_path": (
            f"notebooks/phase2/{filename}"
        ),
        "roadmap_line": unit["start_line"],
        "roadmap_section_sha256": unit["section_sha256"],
    }

assert len(P2AUD_C011_EXPECTED) == 24
assert len({
    item["expected_filename"]
    for item in P2AUD_C011_EXPECTED.values()
}) == 24


def p2aud_c011_path_role(relative_path):
    parts = {
        part.casefold()
        for part in Path(relative_path).parts
    }

    if "audit" in parts:
        return "AUDIT"

    if parts.intersection({
        "archive",
        "archives",
        "archived",
        "old",
        "backup",
        "backups",
    }):
        return "ARCHIVE_OR_BACKUP"

    if Path(relative_path).parent == Path(
        "notebooks/phase2"
    ):
        return "PHASE2_ROOT"

    return "PHASE2_SUBDIRECTORY"


def p2aud_c011_notebook_prefix(filename):
    match = re.match(
        r"^(?:NB)?(\d{2})(?=[A-Za-z_.-]|$)",
        filename,
        re.IGNORECASE,
    )

    if not match:
        return None

    number = int(match.group(1))

    if 0 <= number <= 23:
        return f"NB{number:02d}"

    return None


def p2aud_c011_cell_source(cell):
    source = cell.get("source", "")

    if isinstance(source, list):
        return "".join(
            part for part in source
            if isinstance(part, str)
        )

    if isinstance(source, str):
        return source

    return ""


def p2aud_c011_inspect_notebook(path):
    relative_path = path.relative_to(
        P2AUD_C011_ROOT
    ).as_posix()

    record = {
        "relative_path": relative_path,
        "filename": path.name,
        "path_role": p2aud_c011_path_role(
            relative_path
        ),
        "notebook_prefix": (
            p2aud_c011_notebook_prefix(path.name)
        ),
        "read_status": "NOT_READ",
        "sha256": None,
        "size_bytes": None,
        "nbformat": None,
        "total_cells": None,
        "code_cells": None,
        "markdown_cells": None,
        "other_cells": None,
        "saved_unrun_code_cells": None,
        "saved_error_outputs": None,
        "code_cells_with_blank_first_line": None,
        "code_cells_with_id_header": None,
        "read_error_type": None,
    }

    try:
        size = path.stat().st_size
        record["size_bytes"] = size

        if size > P2AUD_C011_MAX_NOTEBOOK_BYTES:
            record["read_status"] = "SKIPPED_SIZE_LIMIT"
            return record

        raw = path.read_bytes()
        record["sha256"] = hashlib.sha256(
            raw
        ).hexdigest()

        notebook = json.loads(
            raw.decode("utf-8-sig")
        )

        if not isinstance(notebook, dict):
            raise ValueError("Notebook root is not an object")

        cells = notebook.get("cells")

        if not isinstance(cells, list):
            raise ValueError("Notebook cells are not a list")

        counts = Counter()

        for cell in cells:
            if not isinstance(cell, dict):
                raise ValueError("Invalid notebook cell")

            cell_type = cell.get("cell_type")

            if cell_type == "markdown":
                counts["markdown"] += 1

            elif cell_type == "code":
                counts["code"] += 1

                source = p2aud_c011_cell_source(cell)

                if (
                    source.strip()
                    and cell.get("execution_count") is None
                ):
                    counts["saved_unrun"] += 1

                if source.strip():
                    first_line = source.split("\n", 1)[0]

                    if not first_line.strip():
                        counts["blank_first_line"] += 1

                    if re.match(
                        r"^#\s*(?:NB\d{2}|P2AUD)"
                        r"[-_ ]?C\d{2,3}\b",
                        first_line,
                        re.IGNORECASE,
                    ):
                        counts["id_header"] += 1

                outputs = cell.get("outputs", [])

                if isinstance(outputs, list):
                    counts["errors"] += sum(
                        isinstance(output, dict)
                        and output.get("output_type") == "error"
                        for output in outputs
                    )

            else:
                counts["other"] += 1

        record.update({
            "read_status": "READ_SUCCESS",
            "nbformat": notebook.get("nbformat"),
            "total_cells": len(cells),
            "code_cells": counts["code"],
            "markdown_cells": counts["markdown"],
            "other_cells": counts["other"],
            "saved_unrun_code_cells": counts["saved_unrun"],
            "saved_error_outputs": counts["errors"],
            "code_cells_with_blank_first_line": (
                counts["blank_first_line"]
            ),
            "code_cells_with_id_header": (
                counts["id_header"]
            ),
        })

    except (OSError, ValueError, TypeError) as exc:
        record["read_status"] = "READ_ERROR"
        record["read_error_type"] = type(exc).__name__

    return record


P2AUD_C011_DISCOVERED_PATHS = sorted(
    (
        path
        for path in P2AUD_C011_NOTEBOOK_DIR.rglob(
            "*.ipynb"
        )
        if path.is_file()
        and ".ipynb_checkpoints" not in path.parts
        and path.resolve().is_relative_to(
            P2AUD_C011_ROOT
        )
    ),
    key=lambda path: path.as_posix().casefold(),
)

assert len(P2AUD_C011_DISCOVERED_PATHS) <= (
    P2AUD_C011_MAX_FILES
), "Notebook inventory limit exceeded"

P2AUD_C011_NOTEBOOK_FILES = [
    p2aud_c011_inspect_notebook(path)
    for path in P2AUD_C011_DISCOVERED_PATHS
]

P2AUD_C011_FILES_BY_NAME = defaultdict(list)

for record in P2AUD_C011_NOTEBOOK_FILES:
    P2AUD_C011_FILES_BY_NAME[
        record["filename"]
    ].append(record)

P2AUD_C011_NOTEBOOK_REGISTRY = {}

for notebook_id, planned in (
    P2AUD_C011_EXPECTED.items()
):
    expected_path = planned[
        "expected_relative_path"
    ]

    matches = list(
        P2AUD_C011_FILES_BY_NAME.get(
            planned["expected_filename"],
            [],
        )
    )

    direct_matches = [
        record
        for record in matches
        if record["relative_path"] == expected_path
    ]

    if direct_matches:
        selected = direct_matches[0]
        discovery_status = "EXACT_PLANNED_PATH_FOUND"

    elif len(matches) == 1:
        selected = matches[0]
        discovery_status = "FOUND_AT_DIFFERENT_PATH"

    elif len(matches) > 1:
        selected = None
        discovery_status = "AMBIGUOUS_MULTIPLE_PATHS"

    else:
        selected = None
        discovery_status = "NOT_FOUND_BY_PLANNED_FILENAME"

    P2AUD_C011_NOTEBOOK_REGISTRY[
        notebook_id
    ] = {
        **planned,
        "discovery_status": discovery_status,
        "matching_file_count": len(matches),
        "matching_paths": [
            item["relative_path"]
            for item in matches
        ],
        "selected_path": (
            selected["relative_path"]
            if selected is not None
            else None
        ),
        "selected_file_sha256": (
            selected["sha256"]
            if selected is not None
            else None
        ),
        "selected_read_status": (
            selected["read_status"]
            if selected is not None
            else "NOT_SELECTED"
        ),
        "saved_code_cells": (
            selected["code_cells"]
            if selected is not None
            else None
        ),
        "saved_unrun_code_cells": (
            selected["saved_unrun_code_cells"]
            if selected is not None
            else None
        ),
        "saved_error_outputs": (
            selected["saved_error_outputs"]
            if selected is not None
            else None
        ),
        "completion_status": "NOT_EVALUATED",
        "formal_gate_status": "NOT_EVALUATED",
    }


P2AUD_C011_EXPECTED_FILENAMES = {
    item["expected_filename"]
    for item in P2AUD_C011_EXPECTED.values()
}

P2AUD_C011_VARIANT_REGISTRY = [
    {
        "relative_path": record["relative_path"],
        "filename": record["filename"],
        "path_role": record["path_role"],
        "notebook_prefix": record["notebook_prefix"],
        "read_status": record["read_status"],
        "sha256": record["sha256"],
        "code_cells": record["code_cells"],
        "saved_unrun_code_cells": (
            record["saved_unrun_code_cells"]
        ),
        "saved_error_outputs": (
            record["saved_error_outputs"]
        ),
        "relation_to_planned": (
            "PLANNED_FILENAME_ALTERNATE_LOCATION"
            if record["filename"]
            in P2AUD_C011_EXPECTED_FILENAMES
            else "NON_PLANNED_FILENAME"
        ),
        "acceptance_status": "NOT_EVALUATED",
    }
    for record in P2AUD_C011_NOTEBOOK_FILES
    if (
        record["filename"]
        not in P2AUD_C011_EXPECTED_FILENAMES
        or record["path_role"] != "PHASE2_ROOT"
    )
]

P2AUD_C011_DISCOVERY_COUNTS = Counter(
    record["discovery_status"]
    for record in P2AUD_C011_NOTEBOOK_REGISTRY.values()
)

P2AUD_C011_READ_COUNTS = Counter(
    record["read_status"]
    for record in P2AUD_C011_NOTEBOOK_FILES
)

P2AUD_C011_ROLE_COUNTS = Counter(
    record["path_role"]
    for record in P2AUD_C011_NOTEBOOK_FILES
)

P2AUD_C011_FOCUS = {
    "P2AUD-RV-001": ["NB19", "NB21"],
    "P2AUD-RV-002": [
        "NB01", "NB02", "NB03", "NB04",
        "NB05", "NB06", "NB07",
    ],
    "P2AUD-RV-003": ["NB11", "NB12"],
    "P2AUD-RV-004": ["NB10"],
    "P2AUD-RV-005": ["NB22", "NB23"],
}

P2AUD_C011_FOCUS_SUMMARY = {}

for review_id, notebook_ids in P2AUD_C011_FOCUS.items():
    P2AUD_C011_FOCUS_SUMMARY[review_id] = [
        {
            "notebook_id": notebook_id,
            "discovery_status": (
                P2AUD_C011_NOTEBOOK_REGISTRY[
                    notebook_id
                ]["discovery_status"]
            ),
            "selected_path": (
                P2AUD_C011_NOTEBOOK_REGISTRY[
                    notebook_id
                ]["selected_path"]
            ),
            "read_status": (
                P2AUD_C011_NOTEBOOK_REGISTRY[
                    notebook_id
                ]["selected_read_status"]
            ),
        }
        for notebook_id in notebook_ids
    ]

P2AUD_C011_UNRESOLVED = [
    notebook_id
    for notebook_id, record in (
        P2AUD_C011_NOTEBOOK_REGISTRY.items()
    )
    if (
        record["discovery_status"]
        != "EXACT_PLANNED_PATH_FOUND"
        or record["selected_read_status"]
        != "READ_SUCCESS"
    )
]

P2AUD_C011_REPORT = {
    "schema_version": "P2AUD_ACTUAL_NOTEBOOK_INVENTORY_V1",
    "cell_id": "P2AUD-C011",
    "action": 11,
    "status": (
        "P2AUD_C011_INVENTORY_COMPLETE"
        if not P2AUD_C011_UNRESOLVED
        else "P2AUD_C011_INVENTORY_COMPLETE_WITH_REVIEW_FLAGS"
    ),
    "expected_notebooks": 24,
    "planned_notebooks_accounted_for": len(
        P2AUD_C011_NOTEBOOK_REGISTRY
    ),
    "total_ipynb_files_discovered": len(
        P2AUD_C011_NOTEBOOK_FILES
    ),
    "discovery_status_counts": dict(
        sorted(P2AUD_C011_DISCOVERY_COUNTS.items())
    ),
    "read_status_counts": dict(
        sorted(P2AUD_C011_READ_COUNTS.items())
    ),
    "file_role_counts": dict(
        sorted(P2AUD_C011_ROLE_COUNTS.items())
    ),
    "additional_or_alternate_files": len(
        P2AUD_C011_VARIANT_REGISTRY
    ),
    "unresolved_planned_notebooks": (
        P2AUD_C011_UNRESOLVED
    ),
    "notebook_summary": [
        {
            "notebook_id": notebook_id,
            "expected_filename": item["expected_filename"],
            "discovery_status": item["discovery_status"],
            "selected_path": item["selected_path"],
            "read_status": item["selected_read_status"],
            "saved_code_cells": item["saved_code_cells"],
            "saved_unrun_code_cells": (
                item["saved_unrun_code_cells"]
            ),
            "saved_error_outputs": (
                item["saved_error_outputs"]
            ),
        }
        for notebook_id, item in (
            P2AUD_C011_NOTEBOOK_REGISTRY.items()
        )
    ],
    "variant_examples": (
        P2AUD_C011_VARIANT_REGISTRY[:20]
    ),
    "focus_summary": P2AUD_C011_FOCUS_SUMMARY,
    "notebooks_formally_accepted": 0,
    "requirements_implementation_evaluated": 0,
    "formal_gates_passed": 0,
    "files_written": 0,
    "new_api_calls": 0,
    "git_operations": 0,
    "next_cell": "P2AUD-C012",
}

assert P2AUD_C011_REPORT[
    "planned_notebooks_accounted_for"
] == 24

assert P2AUD_C011_REPORT[
    "notebooks_formally_accepted"
] == 0

print(
    json.dumps(
        P2AUD_C011_REPORT,
        ensure_ascii=False,
        indent=2,
    )
)
````

## اقدام ۱۲ — ممیزی شواهد ذخیره‌شده و موجودی Artifactهای فاز دوم

تکمیل ممیزی موجودی C011 با بررسی شواهد واقعی ذخیره‌شده در نوت‌بوک‌های اصلی و فایل‌های مرجع Release، Contract و Test.

- هیچ Gate جدیدی PASS نمی‌شود.

### خروجی‌ها

**حالت اجرا:** READ_ONLY

````python
# P2AUD-C012 — Inspect saved notebook evidence and artifact locations
# Action 12 | READ_ONLY | NO_API | NO_FILE_WRITES | NO_GIT_WRITES

assert P2AUD_C011_REPORT["status"] in {
    "P2AUD_C011_INVENTORY_COMPLETE",
    "P2AUD_C011_INVENTORY_COMPLETE_WITH_REVIEW_FLAGS",
}
assert len(P2AUD_C011_NOTEBOOK_REGISTRY) == 24
assert len(P2AUD_C010_CASE_DOSSIERS) == 5

P2AUD_C012_ROOT = Path(P2AUD_REPO_ROOT).resolve()
P2AUD_C012_MAX_NOTEBOOK_BYTES = 64 * 1024 * 1024
P2AUD_C012_MAX_ARTIFACT_HASH_BYTES = 4 * 1024 * 1024

P2AUD_C012_PATTERNS = {
    "P2AUD-RV-001": (
        r"citation.{0,40}coverage|"
        r"claim.{0,50}citation|"
        r"citation.{0,50}claim|"
        r"citation.{0,30}correctness"
    ),
    "P2AUD-RV-002": (
        r"schema|contract|lineage|"
        r"stable[_-]?id|migration"
    ),
    "P2AUD-RV-003": (
        r"golden|chunk|locator|"
        r"annotation|gold[_-]?documents"
    ),
    "P2AUD-RV-004": (
        r"taxonomy|competency|topic|ontology"
    ),
    "P2AUD-RV-005": (
        r"handoff|rag_request|rag_response|"
        r"readiness|backward[_ -]?compatibility|"
        r"release[_ -]?manifest"
    ),
}

P2AUD_C012_COMPILED = {
    review_id: re.compile(pattern, re.IGNORECASE)
    for review_id, pattern in P2AUD_C012_PATTERNS.items()
}

P2AUD_C012_FOCUS_BY_NOTEBOOK = defaultdict(list)

for review_id, notebook_ids in P2AUD_C011_FOCUS.items():
    assert review_id in P2AUD_C012_COMPILED

    for notebook_id in notebook_ids:
        assert notebook_id in P2AUD_C011_NOTEBOOK_REGISTRY
        P2AUD_C012_FOCUS_BY_NOTEBOOK[
            notebook_id
        ].append(review_id)


def p2aud_c012_source(cell):
    value = cell.get("source", "")

    if isinstance(value, str):
        return value

    if isinstance(value, list):
        return "".join(
            part for part in value
            if isinstance(part, str)
        )

    return ""


P2AUD_C012_NOTEBOOK_EVIDENCE = {}
P2AUD_C012_STATIC_CODE_EVIDENCE = []

for notebook_id, registered in sorted(
    P2AUD_C011_NOTEBOOK_REGISTRY.items()
):
    relative_path = registered["selected_path"]

    if (
        not relative_path
        or registered["selected_read_status"] != "READ_SUCCESS"
    ):
        raise ValueError(
            f"Notebook is not readable: {notebook_id}"
        )

    path = (P2AUD_C012_ROOT / relative_path).resolve()

    assert path.is_relative_to(P2AUD_C012_ROOT)
    assert path.is_file()

    if path.stat().st_size > P2AUD_C012_MAX_NOTEBOOK_BYTES:
        raise ValueError(
            f"Notebook exceeds read limit: {notebook_id}"
        )

    raw = path.read_bytes()
    current_sha256 = hashlib.sha256(raw).hexdigest()

    assert current_sha256 == registered[
        "selected_file_sha256"
    ], f"Notebook changed after C011: {notebook_id}"

    notebook = json.loads(raw.decode("utf-8-sig"))
    cells = notebook.get("cells")

    assert isinstance(cells, list)

    unrun_cell_numbers = []
    error_records = []
    code_count = 0

    code_matches = {
        review_id: {
            "matched_code_cells": 0,
            "keyword_line_matches": 0,
            "examples": [],
        }
        for review_id in P2AUD_C012_FOCUS_BY_NOTEBOOK[
            notebook_id
        ]
    }

    for cell_number, cell in enumerate(cells, start=1):
        if cell.get("cell_type") != "code":
            continue

        code_count += 1
        source = p2aud_c012_source(cell)

        if (
            source.strip()
            and cell.get("execution_count") is None
        ):
            unrun_cell_numbers.append(cell_number)

        outputs = cell.get("outputs", [])

        if not isinstance(outputs, list):
            outputs = []

        for output in outputs:
            if (
                isinstance(output, dict)
                and output.get("output_type") == "error"
            ):
                error_records.append({
                    "cell_number": cell_number,
                    "error_type": str(
                        output.get("ename", "UNKNOWN")
                    )[:100],
                    "execution_count": cell.get(
                        "execution_count"
                    ),
                })

        if not source.strip():
            continue

        source_lines = source.splitlines()

        for review_id, evidence in code_matches.items():
            regex = P2AUD_C012_COMPILED[review_id]
            cell_has_match = False

            for line_number, line in enumerate(
                source_lines, start=1
            ):
                match = regex.search(line)

                if not match:
                    continue

                cell_has_match = True
                evidence["keyword_line_matches"] += 1

                if len(evidence["examples"]) < 3:
                    evidence["examples"].append({
                        "cell_number": cell_number,
                        "line_in_cell": line_number,
                        "matched_term": match.group(0)[:60],
                    })

            if cell_has_match:
                evidence["matched_code_cells"] += 1

    assert code_count == registered["saved_code_cells"], (
        f"Code-cell count changed: {notebook_id}"
    )

    assert len(unrun_cell_numbers) == registered[
        "saved_unrun_code_cells"
    ], f"Unrun count changed: {notebook_id}"

    assert len(error_records) == registered[
        "saved_error_outputs"
    ], f"Error count changed: {notebook_id}"

    P2AUD_C012_NOTEBOOK_EVIDENCE[notebook_id] = {
        "relative_path": relative_path,
        "sha256": current_sha256,
        "code_cells": code_count,
        "saved_unrun_count": len(unrun_cell_numbers),
        "unrun_cell_examples": unrun_cell_numbers[:12],
        "saved_error_count": len(error_records),
        "error_records": error_records[:12],
        "execution_proof": "NOT_VERIFIED_BY_STATIC_INSPECTION",
        "formal_acceptance": "NOT_EVALUATED",
    }

    for review_id, evidence in code_matches.items():
        P2AUD_C012_STATIC_CODE_EVIDENCE.append({
            "review_id": review_id,
            "notebook_id": notebook_id,
            "notebook_path": relative_path,
            "notebook_sha256": current_sha256,
            **evidence,
            "evidence_status": (
                "STATIC_CODE_TOKEN_HITS_NOT_EXECUTION_PROOF"
            ),
        })


# Check exact roadmap-defined artifact locations.
P2AUD_C012_ARTIFACT_TARGETS = {
    "RAG_REQUEST_SCHEMA": (
        "schemas/phase2/rag_request.schema.json", "FILE"
    ),
    "RAG_RESPONSE_SCHEMA": (
        "schemas/phase2/rag_response.schema.json", "FILE"
    ),
    "PHASE3_READINESS_REPORT": (
        "reports/phase2/release/phase3_readiness.md", "FILE"
    ),
    "PHASE3_REQUEST_TEST": (
        "tests/phase2/contracts/test_phase3_request.py", "FILE"
    ),
    "PHASE3_RESPONSE_TEST": (
        "tests/phase2/contracts/test_phase3_response.py", "FILE"
    ),
    "VERSION_TRACE_TEST": (
        "tests/phase2/contracts/test_version_trace.py", "FILE"
    ),
    "DATA_MANIFESTS": (
        "data/phase2/manifests", "DIRECTORY"
    ),
    "DATA_PROCESSED": (
        "data/phase2/processed", "DIRECTORY"
    ),
    "CHUNK_ARTIFACTS": (
        "data/phase2/chunks", "DIRECTORY"
    ),
    "INDEX_ARTIFACTS": (
        "data/phase2/index", "DIRECTORY"
    ),
    "SCHEMAS": (
        "schemas/phase2", "DIRECTORY"
    ),
    "CONFIG": (
        "config/phase2", "DIRECTORY"
    ),
    "REPORTS": (
        "reports/phase2", "DIRECTORY"
    ),
    "CHECKPOINTS": (
        "artifacts/phase2/checkpoints", "DIRECTORY"
    ),
    "RELEASES": (
        "artifacts/phase2/releases", "DIRECTORY"
    ),
    "CONTRACT_TESTS": (
        "tests/phase2/contracts", "DIRECTORY"
    ),
}

P2AUD_C012_ARTIFACT_INVENTORY = {}

for artifact_id, (relative_path, expected_type) in (
    P2AUD_C012_ARTIFACT_TARGETS.items()
):
    path = P2AUD_C012_ROOT / relative_path

    record = {
        "artifact_id": artifact_id,
        "relative_path": relative_path,
        "expected_type": expected_type,
        "status": "NOT_FOUND_AT_PLANNED_PATH",
        "size_bytes": None,
        "sha256": None,
        "direct_child_examples": [],
        "direct_children_sampled": 0,
        "review_status": "NOT_EVALUATED",
    }

    if path.is_symlink():
        record["status"] = "SYMLINK_REQUIRES_REVIEW"

    elif expected_type == "FILE" and path.is_file():
        try:
            size = path.stat().st_size
            record["size_bytes"] = size

            if size <= P2AUD_C012_MAX_ARTIFACT_HASH_BYTES:
                record["sha256"] = hashlib.sha256(
                    path.read_bytes()
                ).hexdigest()
                record["status"] = "FILE_FOUND_HASHED"
            else:
                record["status"] = "FILE_FOUND_HASH_SKIPPED_SIZE"

        except OSError as exc:
            record["status"] = "FILE_READ_ERROR"
            record["error_type"] = type(exc).__name__

    elif expected_type == "DIRECTORY" and path.is_dir():
        record["status"] = "DIRECTORY_FOUND"

        try:
            with os.scandir(path) as entries:
                for index, entry in enumerate(entries):
                    if index >= 40:
                        record["status"] = (
                            "DIRECTORY_FOUND_LISTING_TRUNCATED"
                        )
                        break

                    record["direct_children_sampled"] += 1

                    if len(record["direct_child_examples"]) < 8:
                        record["direct_child_examples"].append({
                            "name": entry.name[:120],
                            "kind": (
                                "DIRECTORY"
                                if entry.is_dir(follow_symlinks=False)
                                else "FILE_OR_OTHER"
                            ),
                        })

        except OSError as exc:
            record["status"] = "DIRECTORY_LIST_ERROR"
            record["error_type"] = type(exc).__name__

    elif path.exists():
        record["status"] = "WRONG_OBJECT_TYPE"

    P2AUD_C012_ARTIFACT_INVENTORY[
        artifact_id
    ] = record


# Release IDs are not guessed. Scan only immediate release folders.
P2AUD_C012_RELEASE_INVENTORY = {
    "release_root": "artifacts/phase2/releases",
    "status": "NOT_FOUND",
    "release_directories_scanned": 0,
    "scan_truncated": False,
    "handoff_contracts_found": [],
    "release_directory_examples": [],
}

release_root = (
    P2AUD_C012_ROOT / "artifacts/phase2/releases"
)

if release_root.is_dir() and not release_root.is_symlink():
    P2AUD_C012_RELEASE_INVENTORY["status"] = "SCANNED"

    try:
        with os.scandir(release_root) as entries:
            for index, entry in enumerate(entries):
                if index >= 100:
                    P2AUD_C012_RELEASE_INVENTORY[
                        "scan_truncated"
                    ] = True
                    break

                if not entry.is_dir(follow_symlinks=False):
                    continue

                P2AUD_C012_RELEASE_INVENTORY[
                    "release_directories_scanned"
                ] += 1

                if len(
                    P2AUD_C012_RELEASE_INVENTORY[
                        "release_directory_examples"
                    ]
                ) < 12:
                    P2AUD_C012_RELEASE_INVENTORY[
                        "release_directory_examples"
                    ].append(entry.name[:120])

                contract_path = (
                    Path(entry.path)
                    / "phase3_handoff_contract.json"
                )

                if (
                    not contract_path.is_symlink()
                    and contract_path.is_file()
                ):
                    P2AUD_C012_RELEASE_INVENTORY[
                        "handoff_contracts_found"
                    ].append(
                        contract_path.relative_to(
                            P2AUD_C012_ROOT
                        ).as_posix()
                    )

    except OSError as exc:
        P2AUD_C012_RELEASE_INVENTORY[
            "status"
        ] = "SCAN_ERROR"
        P2AUD_C012_RELEASE_INVENTORY[
            "error_type"
        ] = type(exc).__name__


P2AUD_C012_NOTEBOOK_FLAGS = {
    notebook_id: {
        "saved_unrun_count": item["saved_unrun_count"],
        "unrun_cell_examples": item["unrun_cell_examples"],
        "saved_error_count": item["saved_error_count"],
        "error_records": item["error_records"],
    }
    for notebook_id, item in (
        P2AUD_C012_NOTEBOOK_EVIDENCE.items()
    )
    if (
        item["saved_unrun_count"] > 0
        or item["saved_error_count"] > 0
    )
}

P2AUD_C012_ARTIFACT_STATUS_COUNTS = Counter(
    record["status"]
    for record in P2AUD_C012_ARTIFACT_INVENTORY.values()
)

P2AUD_C012_CODE_EVIDENCE_SUMMARY = []

for review_id in sorted(P2AUD_C012_COMPILED):
    matches = [
        item
        for item in P2AUD_C012_STATIC_CODE_EVIDENCE
        if item["review_id"] == review_id
    ]

    P2AUD_C012_CODE_EVIDENCE_SUMMARY.append({
        "review_id": review_id,
        "notebooks_examined": len(matches),
        "notebooks_with_token_hits": [
            item["notebook_id"]
            for item in matches
            if item["keyword_line_matches"] > 0
        ],
        "total_keyword_line_matches": sum(
            item["keyword_line_matches"]
            for item in matches
        ),
        "sample_locations": [
            {
                "notebook_id": item["notebook_id"],
                **example,
            }
            for item in matches
            for example in item["examples"][:1]
        ][:8],
        "implementation_verified": False,
    })


P2AUD_C012_REPORT = {
    "schema_version": (
        "P2AUD_STATIC_EXECUTION_AND_ARTIFACT_EVIDENCE_V1"
    ),
    "cell_id": "P2AUD-C012",
    "action": 12,
    "status": "P2AUD_C012_EVIDENCE_STAGED_REVIEW_REQUIRED",
    "planned_notebooks_reverified": len(
        P2AUD_C012_NOTEBOOK_EVIDENCE
    ),
    "notebooks_with_saved_execution_flags": len(
        P2AUD_C012_NOTEBOOK_FLAGS
    ),
    "saved_execution_flags": P2AUD_C012_NOTEBOOK_FLAGS,
    "code_evidence_summary": (
        P2AUD_C012_CODE_EVIDENCE_SUMMARY
    ),
    "planned_artifact_paths_checked": len(
        P2AUD_C012_ARTIFACT_INVENTORY
    ),
    "artifact_status_counts": dict(
        sorted(P2AUD_C012_ARTIFACT_STATUS_COUNTS.items())
    ),
    "artifact_file_status": {
        artifact_id: {
            "status": record["status"],
            "relative_path": record["relative_path"],
            "sha256": record["sha256"],
        }
        for artifact_id, record in (
            P2AUD_C012_ARTIFACT_INVENTORY.items()
        )
        if record["expected_type"] == "FILE"
    },
    "release_inventory": P2AUD_C012_RELEASE_INVENTORY,
    "additional_notebooks_from_c011": len(
        P2AUD_C011_VARIANT_REGISTRY
    ),
    "additional_notebooks_formally_evaluated": 0,
    "requirements_implementation_accepted": 0,
    "final_decisions_approved": 0,
    "formal_gates_passed": 0,
    "files_written": 0,
    "new_api_calls": 0,
    "git_operations": 0,
    "next_cell": "P2AUD-C013",
}

assert P2AUD_C012_REPORT[
    "planned_notebooks_reverified"
] == 24

assert P2AUD_C012_REPORT[
    "planned_artifact_paths_checked"
] == 16

assert P2AUD_C012_REPORT[
    "final_decisions_approved"
] == 0

print(
    json.dumps(
        P2AUD_C012_REPORT,
        ensure_ascii=False,
        indent=2,
    )
)
````

### شرط پایان

هر شش مسیر یافت‌نشده بررسی جایگزین شده باشند، هر ۲۶ فایل اضافی تعیین وضعیت اولیه شده باشد و ارجاعات مربوط به Release و Handoff مستند شوند.

**حالت:** READ_ONLY

**API:** صفر

**Git و عملیات نوشتنی:** صفر

````python
# P2AUD-C013 — Reconcile missing artifacts and notebook variants
# Action 13 | READ_ONLY | DRIFT_AWARE | NO_API | NO_GIT_WRITES

assert P2AUD_C012_REPORT["status"] == (
    "P2AUD_C012_EVIDENCE_STAGED_REVIEW_REQUIRED"
)
assert len(P2AUD_C011_VARIANT_REGISTRY) == 26

P2AUD_C013_ROOT = Path(P2AUD_REPO_ROOT).resolve()
P2AUD_C013_MAX_BYTES = 64 * 1024 * 1024

P2AUD_C013_MISSING_TARGETS = {
    artifact_id: record
    for artifact_id, record in P2AUD_C012_ARTIFACT_INVENTORY.items()
    if record["status"] == "NOT_FOUND_AT_PLANNED_PATH"
}

assert len(P2AUD_C013_MISSING_TARGETS) == 6

P2AUD_C013_SEARCH_ROOTS = (
    "reports",
    "artifacts",
    "tests",
    "schemas",
    "src/phase2",
    "notebooks/phase2",
    "data/phase2",
    "config/phase2",
    "docs/phase2",
)

P2AUD_C013_SKIP_DIRECTORIES = {
    ".git", ".venv", "venv", "node_modules",
    "__pycache__", ".ipynb_checkpoints",
    ".pytest_cache", ".mypy_cache",
    "raw", "interim", "processed",
    "chunks", "embeddings", "indexes",
    "qdrant_storage",
}

P2AUD_C013_SCANNED_FILES = []
P2AUD_C013_SCANNED_DIRECTORIES = []
P2AUD_C013_SCAN_ROOT_STATUS = []
P2AUD_C013_SCAN_TRUNCATED = False

MAX_FILES = 30000
MAX_DIRS = 10000

for relative_root in P2AUD_C013_SEARCH_ROOTS:
    search_root = P2AUD_C013_ROOT / relative_root

    if search_root.is_symlink():
        P2AUD_C013_SCAN_ROOT_STATUS.append({
            "root": relative_root,
            "status": "SKIPPED_SYMLINK",
        })
        continue

    if not search_root.is_dir():
        P2AUD_C013_SCAN_ROOT_STATUS.append({
            "root": relative_root,
            "status": "NOT_FOUND",
        })
        continue

    P2AUD_C013_SCAN_ROOT_STATUS.append({
        "root": relative_root,
        "status": "SCANNED",
    })

    for current_dir, subdirs, filenames in os.walk(
        search_root,
        topdown=True,
        followlinks=False,
    ):
        current = Path(current_dir)

        subdirs[:] = sorted(
            name
            for name in subdirs
            if name not in P2AUD_C013_SKIP_DIRECTORIES
            and not (current / name).is_symlink()
        )

        relative_dir = current.relative_to(
            P2AUD_C013_ROOT
        ).as_posix()

        P2AUD_C013_SCANNED_DIRECTORIES.append(
            relative_dir
        )

        for filename in sorted(filenames):
            file_path = current / filename

            if file_path.is_symlink() or not file_path.is_file():
                continue

            P2AUD_C013_SCANNED_FILES.append(
                file_path.relative_to(
                    P2AUD_C013_ROOT
                ).as_posix()
            )

            if len(P2AUD_C013_SCANNED_FILES) >= MAX_FILES:
                P2AUD_C013_SCAN_TRUNCATED = True
                break

        if (
            P2AUD_C013_SCAN_TRUNCATED
            or len(P2AUD_C013_SCANNED_DIRECTORIES) >= MAX_DIRS
        ):
            P2AUD_C013_SCAN_TRUNCATED = True
            break

    if P2AUD_C013_SCAN_TRUNCATED:
        break


# Locate missing planned artifacts by exact filename or directory name.

P2AUD_C013_ALTERNATIVE_LOCATIONS = {}

for artifact_id, missing_record in sorted(
    P2AUD_C013_MISSING_TARGETS.items()
):
    expected_path = missing_record["relative_path"]
    expected_name = Path(expected_path).name.casefold()
    expected_type = missing_record["expected_type"]

    candidates = (
        P2AUD_C013_SCANNED_FILES
        if expected_type == "FILE"
        else P2AUD_C013_SCANNED_DIRECTORIES
    )

    matches = [
        relative_path
        for relative_path in candidates
        if (
            Path(relative_path).name.casefold() == expected_name
            and relative_path != expected_path
        )
    ]

    P2AUD_C013_ALTERNATIVE_LOCATIONS[artifact_id] = {
        "expected_path": expected_path,
        "expected_type": expected_type,
        "exact_name_alternatives_count": len(matches),
        "exact_name_alternatives": matches[:20],
        "search_status": (
            "ALTERNATIVE_NAME_MATCH_REQUIRES_VALIDATION"
            if matches
            else (
                "SCAN_INCOMPLETE"
                if P2AUD_C013_SCAN_TRUNCATED
                else "NOT_FOUND_IN_BOUNDED_SCAN"
            )
        ),
        "content_validated": False,
        "formal_acceptance": "NOT_EVALUATED",
    }


P2AUD_C013_RELEASE_PATTERN = re.compile(
    r"release|readiness|handoff|phase.?3|version.?trace",
    re.IGNORECASE,
)

P2AUD_C013_RELEASE_FILE_CANDIDATES = [
    relative_path
    for relative_path in P2AUD_C013_SCANNED_FILES
    if P2AUD_C013_RELEASE_PATTERN.search(
        Path(relative_path).name
    )
    and Path(relative_path).suffix.lower() in {
        ".json", ".md", ".yaml", ".yml", ".py", ".txt",
    }
]

P2AUD_C013_RELEASE_DIR_CANDIDATES = [
    relative_path
    for relative_path in P2AUD_C013_SCANNED_DIRECTORIES
    if re.search(
        r"release|handoff|readiness",
        Path(relative_path).name,
        re.IGNORECASE,
    )
]


# Re-fingerprint variant notebooks without modifying them.
# A changed hash is recorded, not treated as an automatic failure.

def p2aud_c013_read_snapshot(relative_path):
    path = P2AUD_C013_ROOT / relative_path

    result = {
        "read_status": "NOT_READ",
        "sha256": None,
        "size_bytes": None,
        "raw": None,
    }

    if path.is_symlink():
        result["read_status"] = "SYMLINK_REQUIRES_REVIEW"
        return result

    if not path.is_file():
        result["read_status"] = "FILE_NOT_FOUND"
        return result

    if not path.resolve().is_relative_to(P2AUD_C013_ROOT):
        result["read_status"] = "OUTSIDE_REPOSITORY"
        return result

    try:
        size = path.stat().st_size
        result["size_bytes"] = size

        if size > P2AUD_C013_MAX_BYTES:
            result["read_status"] = "SKIPPED_SIZE_LIMIT"
            return result

        raw = path.read_bytes()

        result["raw"] = raw
        result["sha256"] = hashlib.sha256(raw).hexdigest()
        result["read_status"] = "READ_SUCCESS"

    except OSError as exc:
        result["read_status"] = "READ_ERROR"
        result["error_type"] = type(exc).__name__

    return result


P2AUD_C013_VARIANT_LEDGER = []
P2AUD_C013_VARIANT_FAMILY_COUNTS = Counter()
P2AUD_C013_VARIANT_SNAPSHOTS = {}

for variant in P2AUD_C011_VARIANT_REGISTRY:
    relative_path = variant["relative_path"]
    family_id = variant["notebook_prefix"] or "UNCLASSIFIED"

    snapshot = p2aud_c013_read_snapshot(relative_path)

    baseline_sha = variant["sha256"]
    current_sha = snapshot["sha256"]

    if snapshot["read_status"] != "READ_SUCCESS":
        snapshot_status = snapshot["read_status"]
    elif current_sha == baseline_sha:
        snapshot_status = "MATCH_C011"
    else:
        snapshot_status = "CHANGED_SINCE_C011"

    if variant["path_role"] == "ARCHIVE_OR_BACKUP":
        classification = "ARCHIVED_OR_BACKUP"
    elif variant["path_role"] == "AUDIT":
        classification = "INDEPENDENT_AUDIT"
    elif family_id.startswith("NB"):
        classification = "ACTIVE_LOCATION_VARIANT_CANDIDATE"
    else:
        classification = "AUXILIARY_NOTEBOOK"

    unrun = variant["saved_unrun_code_cells"]
    errors = variant["saved_error_outputs"]

    has_saved_flags = (
        isinstance(unrun, int) and unrun > 0
    ) or (
        isinstance(errors, int) and errors > 0
    )

    P2AUD_C013_VARIANT_SNAPSHOTS[relative_path] = {
        "sha256_from_c011": baseline_sha,
        "sha256_current": current_sha,
        "snapshot_status": snapshot_status,
    }

    P2AUD_C013_VARIANT_LEDGER.append({
        "relative_path": relative_path,
        "family_id": family_id,
        "path_role": variant["path_role"],
        "classification_hint": classification,
        "sha256_from_c011": baseline_sha,
        "sha256_current": current_sha,
        "snapshot_status": snapshot_status,
        "saved_unrun_code_cells": unrun,
        "saved_error_outputs": errors,
        "saved_execution_flags_from_c011": has_saved_flags,
        "current_execution_counts_reverified": False,
        "replacement_for_primary": "NOT_AUTHORIZED",
        "formal_acceptance": "NOT_EVALUATED",
    })

    P2AUD_C013_VARIANT_FAMILY_COUNTS[family_id] += 1

assert len(P2AUD_C013_VARIANT_LEDGER) == 26


# Inspect references in primary NB22/NB23 and relevant NB22 variants.
# Inspect the current observed bytes, explicitly labeling any drift.

P2AUD_C013_REFERENCE_PATTERNS = {
    "HANDOFF_CONTRACT": r"phase3_handoff_contract(?:\.json)?",
    "READINESS_REPORT": r"phase3_readiness(?:\.md)?",
    "REQUEST_TEST": r"test_phase3_request",
    "RESPONSE_TEST": r"test_phase3_response",
    "VERSION_TRACE_TEST": r"test_version_trace",
    "RELEASE_MANIFEST": r"release[_ -]?manifest",
    "FORMAL_READINESS": (
        r"formal_readiness|accepted_for_phase3|phase2_completed"
    ),
    "CONTRACT_TEST_LOGIC": (
        r"pytest|unittest|def test_|contract.tests?"
    ),
}

P2AUD_C013_COMPILED_PATTERNS = {
    label: re.compile(pattern, re.IGNORECASE)
    for label, pattern in P2AUD_C013_REFERENCE_PATTERNS.items()
}

P2AUD_C013_REFERENCE_INPUTS = []

for notebook_id in ("NB22", "NB23"):
    record = P2AUD_C011_NOTEBOOK_REGISTRY[notebook_id]

    P2AUD_C013_REFERENCE_INPUTS.append({
        "notebook_id": notebook_id,
        "relative_path": record["selected_path"],
        "expected_sha256": (
            P2AUD_C012_NOTEBOOK_EVIDENCE[notebook_id]["sha256"]
        ),
        "baseline_cell": "C012",
        "kind": "PRIMARY",
    })

for variant in P2AUD_C011_VARIANT_REGISTRY:
    if (
        variant["notebook_prefix"] == "NB22"
        and variant["path_role"] == "PHASE2_ROOT"
    ):
        P2AUD_C013_REFERENCE_INPUTS.append({
            "notebook_id": "NB22",
            "relative_path": variant["relative_path"],
            "expected_sha256": variant["sha256"],
            "baseline_cell": "C011",
            "kind": "VARIANT",
        })

P2AUD_C013_NOTEBOOK_REFERENCES = []

for reference_input in P2AUD_C013_REFERENCE_INPUTS:
    relative_path = reference_input["relative_path"]
    snapshot = p2aud_c013_read_snapshot(relative_path)

    current_sha = snapshot["sha256"]
    expected_sha = reference_input["expected_sha256"]

    if snapshot["read_status"] != "READ_SUCCESS":
        snapshot_status = snapshot["read_status"]
    elif current_sha == expected_sha:
        snapshot_status = (
            "MATCH_" + reference_input["baseline_cell"]
        )
    else:
        snapshot_status = (
            "CHANGED_SINCE_" + reference_input["baseline_cell"]
        )

    reference_result = {
        "notebook_id": reference_input["notebook_id"],
        "relative_path": relative_path,
        "kind": reference_input["kind"],
        "baseline_cell": reference_input["baseline_cell"],
        "sha256_baseline": expected_sha,
        "sha256_observed": current_sha,
        "snapshot_status": snapshot_status,
        "reference_status": "NOT_SCANNED",
        "matches": {},
        "evidence_status": (
            "STATIC_SOURCE_REFERENCE_NOT_EXECUTION_PROOF"
        ),
    }

    if snapshot["read_status"] != "READ_SUCCESS":
        reference_result["reference_status"] = (
            "SKIPPED_UNREADABLE"
        )
        P2AUD_C013_NOTEBOOK_REFERENCES.append(
            reference_result
        )
        continue

    try:
        notebook = json.loads(
            snapshot["raw"].decode("utf-8-sig")
        )

        cells = notebook.get("cells", [])

        if not isinstance(cells, list):
            raise ValueError("Notebook cells must be a list")

        matches_by_label = {
            label: {
                "line_match_count": 0,
                "examples": [],
            }
            for label in P2AUD_C013_REFERENCE_PATTERNS
        }

        for cell_number, cell in enumerate(
            cells, start=1
        ):
            if not isinstance(cell, dict):
                continue

            if cell.get("cell_type") not in {
                "code", "markdown"
            }:
                continue

            source = cell.get("source", "")

            if isinstance(source, list):
                source = "".join(
                    part for part in source
                    if isinstance(part, str)
                )

            if not isinstance(source, str):
                continue

            for line_number, line in enumerate(
                source.splitlines(), start=1
            ):
                for label, regex in (
                    P2AUD_C013_COMPILED_PATTERNS.items()
                ):
                    match = regex.search(line)

                    if not match:
                        continue

                    evidence = matches_by_label[label]
                    evidence["line_match_count"] += 1

                    if len(evidence["examples"]) < 3:
                        evidence["examples"].append({
                            "cell_number": cell_number,
                            "cell_type": cell.get("cell_type"),
                            "line_in_cell": line_number,
                            "matched_term": match.group(0)[:70],
                        })

        reference_result["matches"] = {
            label: evidence
            for label, evidence in matches_by_label.items()
            if evidence["line_match_count"] > 0
        }

        reference_result["reference_status"] = (
            "STATIC_REFERENCES_EXTRACTED"
        )

    except (ValueError, TypeError, UnicodeError) as exc:
        reference_result["reference_status"] = (
            "NOTEBOOK_PARSE_ERROR"
        )
        reference_result["error_type"] = type(exc).__name__

    P2AUD_C013_NOTEBOOK_REFERENCES.append(
        reference_result
    )


# Preserve snapshot drift as an explicit unresolved audit finding.

P2AUD_C013_VARIANT_DRIFT = [
    {
        "relative_path": record["relative_path"],
        "sha256_from_c011": record["sha256_from_c011"],
        "sha256_current": record["sha256_current"],
        "snapshot_status": record["snapshot_status"],
    }
    for record in P2AUD_C013_VARIANT_LEDGER
    if record["snapshot_status"] != "MATCH_C011"
]

P2AUD_C013_REFERENCE_DRIFT = [
    {
        "relative_path": record["relative_path"],
        "kind": record["kind"],
        "snapshot_status": record["snapshot_status"],
        "sha256_baseline": record["sha256_baseline"],
        "sha256_observed": record["sha256_observed"],
    }
    for record in P2AUD_C013_NOTEBOOK_REFERENCES
    if record["snapshot_status"] not in {
        "MATCH_C011", "MATCH_C012"
    }
]

P2AUD_C013_EXECUTION_FLAGS = [
    record
    for record in P2AUD_C013_VARIANT_LEDGER
    if record["saved_execution_flags_from_c011"]
]

P2AUD_C013_ALT_COUNTS = Counter(
    record["search_status"]
    for record in P2AUD_C013_ALTERNATIVE_LOCATIONS.values()
)

P2AUD_C013_CLASS_COUNTS = Counter(
    record["classification_hint"]
    for record in P2AUD_C013_VARIANT_LEDGER
)

P2AUD_C013_SNAPSHOT_COUNTS = Counter(
    record["snapshot_status"]
    for record in P2AUD_C013_VARIANT_LEDGER
)

P2AUD_C013_HAS_DRIFT = bool(
    P2AUD_C013_VARIANT_DRIFT
    or P2AUD_C013_REFERENCE_DRIFT
)

P2AUD_C013_REPORT = {
    "schema_version": (
        "P2AUD_ARTIFACT_AND_VARIANT_RECONCILIATION_V2"
    ),
    "cell_id": "P2AUD-C013",
    "action": 13,
    "status": (
        "P2AUD_C013_SNAPSHOT_DRIFT_REVIEW_REQUIRED"
        if P2AUD_C013_HAS_DRIFT
        else "P2AUD_C013_ALTERNATIVES_STAGED_NOT_FORMALLY_VALIDATED"
    ),
    "missing_planned_paths_in_c012": len(
        P2AUD_C013_MISSING_TARGETS
    ),
    "scan_roots": P2AUD_C013_SCAN_ROOT_STATUS,
    "files_scanned": len(P2AUD_C013_SCANNED_FILES),
    "directories_scanned": len(
        P2AUD_C013_SCANNED_DIRECTORIES
    ),
    "scan_truncated": P2AUD_C013_SCAN_TRUNCATED,
    "alternative_location_status_counts": dict(
        sorted(P2AUD_C013_ALT_COUNTS.items())
    ),
    "alternative_locations": (
        P2AUD_C013_ALTERNATIVE_LOCATIONS
    ),
    "release_named_file_candidates_count": len(
        P2AUD_C013_RELEASE_FILE_CANDIDATES
    ),
    "release_named_file_examples": (
        P2AUD_C013_RELEASE_FILE_CANDIDATES[:20]
    ),
    "release_named_directory_candidates_count": len(
        P2AUD_C013_RELEASE_DIR_CANDIDATES
    ),
    "release_named_directory_examples": (
        P2AUD_C013_RELEASE_DIR_CANDIDATES[:15]
    ),
    "variant_notebooks_accounted_for": len(
        P2AUD_C013_VARIANT_LEDGER
    ),
    "variant_family_counts": dict(
        sorted(P2AUD_C013_VARIANT_FAMILY_COUNTS.items())
    ),
    "variant_classification_counts": dict(
        sorted(P2AUD_C013_CLASS_COUNTS.items())
    ),
    "variant_snapshot_status_counts": dict(
        sorted(P2AUD_C013_SNAPSHOT_COUNTS.items())
    ),
    "variant_snapshot_drift_count": len(
        P2AUD_C013_VARIANT_DRIFT
    ),
    "variant_snapshot_drift": (
        P2AUD_C013_VARIANT_DRIFT
    ),
    "variants_with_saved_execution_flags": len(
        P2AUD_C013_EXECUTION_FLAGS
    ),
    "variant_execution_flag_examples": [
        {
            "relative_path": record["relative_path"],
            "saved_unrun_code_cells": (
                record["saved_unrun_code_cells"]
            ),
            "saved_error_outputs": (
                record["saved_error_outputs"]
            ),
            "counts_reverified": False,
        }
        for record in P2AUD_C013_EXECUTION_FLAGS[:15]
    ],
    "notebook_reference_files_examined": len(
        P2AUD_C013_NOTEBOOK_REFERENCES
    ),
    "notebook_reference_summary": [
        {
            "relative_path": record["relative_path"],
            "kind": record["kind"],
            "snapshot_status": record["snapshot_status"],
            "reference_status": record["reference_status"],
            "reference_match_counts": {
                label: evidence["line_match_count"]
                for label, evidence in record["matches"].items()
            },
        }
        for record in P2AUD_C013_NOTEBOOK_REFERENCES
    ],
    "reference_snapshot_drift_count": len(
        P2AUD_C013_REFERENCE_DRIFT
    ),
    "reference_snapshot_drift": (
        P2AUD_C013_REFERENCE_DRIFT
    ),
    "snapshot_drift_requires_reconciliation": (
        P2AUD_C013_HAS_DRIFT
    ),
    "primary_notebooks_modified_by_audit": 0,
    "variants_promoted_to_primary": 0,
    "release_formally_accepted": False,
    "requirements_implementation_accepted": 0,
    "formal_gates_passed": 0,
    "files_written": 0,
    "new_api_calls": 0,
    "git_operations": 0,
    "next_cell": "P2AUD-C014",
}

assert P2AUD_C013_REPORT[
    "variant_notebooks_accounted_for"
] == 26

assert P2AUD_C013_REPORT[
    "variants_promoted_to_primary"
] == 0

print(
    json.dumps(
        P2AUD_C013_REPORT,
        ensure_ascii=False,
        indent=2,
    )
)
````

````python
# P2AUD-C014 — Focused artifact search and notebook drift verification
# Action 14 | READ_ONLY | NO_API | NO_FILE_WRITES | NO_GIT_WRITES

assert P2AUD_C013_REPORT["status"] in {
    "P2AUD_C013_SNAPSHOT_DRIFT_REVIEW_REQUIRED",
    "P2AUD_C013_ALTERNATIVES_STAGED_NOT_FORMALLY_VALIDATED",
}
assert len(P2AUD_C013_ALTERNATIVE_LOCATIONS) == 6
assert len(P2AUD_C013_VARIANT_LEDGER) == 26

P2AUD_C014_ROOT = Path(P2AUD_REPO_ROOT).resolve()

# Search each relevant location independently.
# Limits apply per root, avoiding the C013 global 30,000-file cutoff.

P2AUD_C014_ROOT_SPECS = (
    ("reports/phase2/release", 10000, 8),
    ("tests/phase2", 4000, 6),
    ("config", 2000, 5),
    ("schemas/phase2", 2000, 5),
    ("artifacts/phase2", 2000, 1),
)

P2AUD_C014_SKIP = {
    ".git", ".venv", "venv", "__pycache__",
    ".ipynb_checkpoints", "node_modules",
    "raw", "interim", "processed",
    "chunks", "embeddings", "qdrant_storage",
}

P2AUD_C014_SEARCH_RESULTS = []
P2AUD_C014_INDEXED_FILES = set()
P2AUD_C014_INDEXED_DIRS = set()

for relative_root, max_files, max_depth in P2AUD_C014_ROOT_SPECS:
    search_root = P2AUD_C014_ROOT / relative_root

    result = {
        "root": relative_root,
        "status": "NOT_FOUND",
        "files_scanned": 0,
        "directories_scanned": 0,
        "max_files": max_files,
        "max_depth": max_depth,
        "truncated": False,
        "errors": [],
    }

    if search_root.is_symlink():
        result["status"] = "SKIPPED_SYMLINK"
        P2AUD_C014_SEARCH_RESULTS.append(result)
        continue

    if not search_root.is_dir():
        P2AUD_C014_SEARCH_RESULTS.append(result)
        continue

    result["status"] = "SCANNED"

    for current_dir, subdirs, filenames in os.walk(
        search_root,
        topdown=True,
        followlinks=False,
        onerror=lambda error: result["errors"].append(
            type(error).__name__
        ),
    ):
        current = Path(current_dir)
        relative_current = current.relative_to(search_root)
        depth = len(relative_current.parts)

        result["directories_scanned"] += 1

        P2AUD_C014_INDEXED_DIRS.add(
            current.relative_to(
                P2AUD_C014_ROOT
            ).as_posix()
        )

        if depth >= max_depth:
            subdirs[:] = []
        else:
            subdirs[:] = sorted(
                name for name in subdirs
                if name not in P2AUD_C014_SKIP
                and not (current / name).is_symlink()
            )

        for filename in sorted(filenames):
            path = current / filename

            if path.is_symlink() or not path.is_file():
                continue

            relative_path = path.relative_to(
                P2AUD_C014_ROOT
            ).as_posix()

            P2AUD_C014_INDEXED_FILES.add(relative_path)
            result["files_scanned"] += 1

            if result["files_scanned"] >= max_files:
                result["truncated"] = True
                break

        if result["truncated"]:
            break

    if result["errors"]:
        result["status"] = "SCANNED_WITH_ERRORS"
    elif result["truncated"]:
        result["status"] = "SCANNED_WITH_LIMIT"

    P2AUD_C014_SEARCH_RESULTS.append(result)


# Recheck every missing planned path directly.

P2AUD_C014_TARGET_RECHECK = {}

for artifact_id, previous in sorted(
    P2AUD_C013_ALTERNATIVE_LOCATIONS.items()
):
    expected_path = previous["expected_path"]
    expected_type = previous["expected_type"]

    target_path = P2AUD_C014_ROOT / expected_path

    if target_path.is_symlink():
        direct_status = "SYMLINK_REQUIRES_REVIEW"
    elif expected_type == "FILE" and target_path.is_file():
        direct_status = "NOW_FOUND_AT_PLANNED_PATH"
    elif expected_type == "DIRECTORY" and target_path.is_dir():
        direct_status = "NOW_FOUND_AT_PLANNED_PATH"
    else:
        direct_status = "STILL_MISSING_AT_PLANNED_PATH"

    if expected_type == "FILE":
        alternatives = [
            path
            for path in P2AUD_C014_INDEXED_FILES
            if (
                Path(path).name.casefold()
                == Path(expected_path).name.casefold()
                and path != expected_path
            )
        ]

    elif artifact_id == "CONFIG":
        # A directory named "phase2" is not necessarily config/phase2.
        alternatives = [
            path
            for path in P2AUD_C014_INDEXED_DIRS
            if (
                Path(path).name.casefold() == "phase2"
                and Path(path).parent.name.casefold() == "config"
                and path != expected_path
            )
        ]

    elif artifact_id == "RELEASES":
        alternatives = [
            path
            for path in P2AUD_C014_INDEXED_DIRS
            if (
                Path(path).name.casefold() == "releases"
                and Path(path).parent.name.casefold() == "phase2"
                and path != expected_path
            )
        ]

    else:
        alternatives = [
            path
            for path in P2AUD_C014_INDEXED_DIRS
            if (
                Path(path).name.casefold()
                == Path(expected_path).name.casefold()
                and path != expected_path
            )
        ]

    P2AUD_C014_TARGET_RECHECK[artifact_id] = {
        "planned_path": expected_path,
        "expected_type": expected_type,
        "direct_status": direct_status,
        "alternative_exact_matches": sorted(alternatives)[:20],
        "alternative_match_count": len(alternatives),
        "validation_status": "NOT_EVALUATED",
        "formal_acceptance": "NOT_EVALUATED",
        "global_absence_proven": False,
    }


# Inspect selected release/readiness reports by filename.
# Historical report evidence cannot by itself approve a release.

P2AUD_C014_REPORT_FILE_PATTERN = re.compile(
    r"readiness|final.release.gate|handoff|release.notes",
    re.IGNORECASE,
)

report_candidates = {
    path
    for path in P2AUD_C014_INDEXED_FILES
    if path.startswith("reports/phase2/release/")
    and Path(path).suffix.casefold() in {".json", ".md"}
    and P2AUD_C014_REPORT_FILE_PATTERN.search(
        Path(path).name
    )
}

# Include already-observed paths from C013 even when the
# focused directory enumeration has reached its limit.
for path in P2AUD_C013_RELEASE_FILE_CANDIDATES:
    if (
        path.startswith("reports/phase2/release/")
        and Path(path).suffix.casefold() in {".json", ".md"}
        and P2AUD_C014_REPORT_FILE_PATTERN.search(
            Path(path).name
        )
    ):
        report_candidates.add(path)


def p2aud_c014_report_priority(relative_path):
    name = relative_path.casefold()

    if "nb23.c012" in name:
        return (0, name)
    if "final_release_gate" in name:
        return (1, name)
    if "draft_readiness" in name:
        return (2, name)
    if "handoff_result" in name:
        return (3, name)
    if "readiness" in name:
        return (4, name)
    return (5, name)


P2AUD_C014_SELECTED_REPORTS = sorted(
    report_candidates,
    key=p2aud_c014_report_priority,
)[:16]

P2AUD_C014_STATUS_PATTERN = re.compile(
    r"formal_readiness|accepted_for_phase3|"
    r"phase2_completed|quality_gate|release_gate|"
    r"\bBLOCKED\b|\bFAILED\b|\bFAIL\b|"
    r"\bDRAFT\b|\bPASS\b",
    re.IGNORECASE,
)

P2AUD_C014_RELEASE_REPORT_EVIDENCE = []

for relative_path in P2AUD_C014_SELECTED_REPORTS:
    path = P2AUD_C014_ROOT / relative_path

    report_record = {
        "relative_path": relative_path,
        "read_status": "NOT_READ",
        "sha256": None,
        "size_bytes": None,
        "status_marker_count": 0,
        "status_marker_examples": [],
        "report_class": (
            "DRAFT_READINESS_CANDIDATE"
            if "draft" in relative_path.casefold()
            else "RELEASE_OR_HANDOFF_REPORT_CANDIDATE"
        ),
        "formal_acceptance": "NOT_VERIFIED",
    }

    try:
        if path.is_symlink() or not path.is_file():
            report_record["read_status"] = "NOT_READABLE_FILE"
        elif path.stat().st_size > 2_000_000:
            report_record["read_status"] = "SKIPPED_SIZE_LIMIT"
        else:
            raw = path.read_bytes()
            content = raw.decode("utf-8-sig")

            report_record["sha256"] = hashlib.sha256(
                raw
            ).hexdigest()
            report_record["size_bytes"] = len(raw)
            report_record["read_status"] = "READ_SUCCESS"

            for number, line in enumerate(
                content.splitlines(),
                start=1,
            ):
                match = P2AUD_C014_STATUS_PATTERN.search(line)

                if match:
                    report_record["status_marker_count"] += 1

                    if len(
                        report_record["status_marker_examples"]
                    ) < 8:
                        report_record[
                            "status_marker_examples"
                        ].append({
                            "line": number,
                            "matched_term": match.group(0),
                        })

    except (OSError, UnicodeError) as exc:
        report_record["read_status"] = "READ_ERROR"
        report_record["error_type"] = type(exc).__name__

    P2AUD_C014_RELEASE_REPORT_EVIDENCE.append(
        report_record
    )


# Inventory alternative contract-test filenames.
# Presence of a test source file does not prove it passed.

P2AUD_C014_TEST_NAME_PATTERN = re.compile(
    r"contract|phase3|request|response|version|trace",
    re.IGNORECASE,
)

P2AUD_C014_TEST_CANDIDATES = sorted(
    path
    for path in P2AUD_C014_INDEXED_FILES
    if (
        path.startswith("tests/phase2/")
        and path.endswith(".py")
        and P2AUD_C014_TEST_NAME_PATTERN.search(
            Path(path).name
        )
    )
)


# Independently recheck the modified NB22e variant.

P2AUD_C014_DRIFT_PATH = (
    "notebooks/phase2/"
    "22e_retrieval_experiment_lab_v2.ipynb"
)

P2AUD_C014_DRIFT_BASELINE = next(
    item
    for item in P2AUD_C011_VARIANT_REGISTRY
    if item["relative_path"] == P2AUD_C014_DRIFT_PATH
)

P2AUD_C014_DRIFT_C013 = (
    P2AUD_C013_VARIANT_SNAPSHOTS[
        P2AUD_C014_DRIFT_PATH
    ]
)

P2AUD_C014_DRIFT_RECHECK = {
    "relative_path": P2AUD_C014_DRIFT_PATH,
    "sha256_c011": P2AUD_C014_DRIFT_BASELINE["sha256"],
    "sha256_c013": P2AUD_C014_DRIFT_C013["sha256_current"],
    "sha256_c014": None,
    "status": "NOT_READ",
    "code_cells_c011": P2AUD_C014_DRIFT_BASELINE["code_cells"],
    "code_cells_c014": None,
    "unrun_cells_c014": None,
    "error_outputs_c014": None,
    "reason_for_change": "NOT_DETERMINED",
    "execution_validated": False,
}

drift_file = P2AUD_C014_ROOT / P2AUD_C014_DRIFT_PATH

try:
    if drift_file.is_symlink() or not drift_file.is_file():
        P2AUD_C014_DRIFT_RECHECK["status"] = (
            "FILE_UNAVAILABLE"
        )
    elif drift_file.stat().st_size > 64 * 1024 * 1024:
        P2AUD_C014_DRIFT_RECHECK["status"] = (
            "SKIPPED_SIZE_LIMIT"
        )
    else:
        raw = drift_file.read_bytes()
        current_sha = hashlib.sha256(raw).hexdigest()

        notebook = json.loads(raw.decode("utf-8-sig"))
        cells = notebook.get("cells")

        if not isinstance(cells, list):
            raise ValueError("Invalid notebook cells")

        code_cells = [
            cell for cell in cells
            if isinstance(cell, dict)
            and cell.get("cell_type") == "code"
        ]

        unrun_count = sum(
            bool(p2aud_c012_source(cell).strip())
            and cell.get("execution_count") is None
            for cell in code_cells
        )

        error_count = sum(
            isinstance(output, dict)
            and output.get("output_type") == "error"
            for cell in code_cells
            for output in (
                cell.get("outputs", [])
                if isinstance(cell.get("outputs", []), list)
                else []
            )
        )

        P2AUD_C014_DRIFT_RECHECK.update({
            "sha256_c014": current_sha,
            "status": (
                "MATCH_C013_CHANGED_SINCE_C011"
                if current_sha == (
                    P2AUD_C014_DRIFT_C013["sha256_current"]
                )
                and current_sha != (
                    P2AUD_C014_DRIFT_BASELINE["sha256"]
                )
                else (
                    "MATCH_C011"
                    if current_sha == (
                        P2AUD_C014_DRIFT_BASELINE["sha256"]
                    )
                    else "CHANGED_SINCE_PRIOR_SNAPSHOT"
                )
            ),
            "code_cells_c014": len(code_cells),
            "unrun_cells_c014": unrun_count,
            "error_outputs_c014": error_count,
        })

except (OSError, ValueError, TypeError, UnicodeError) as exc:
    P2AUD_C014_DRIFT_RECHECK["status"] = "READ_ERROR"
    P2AUD_C014_DRIFT_RECHECK["error_type"] = type(exc).__name__


P2AUD_C014_TARGET_COUNTS = Counter(
    item["direct_status"]
    for item in P2AUD_C014_TARGET_RECHECK.values()
)

P2AUD_C014_REPORT_READ_COUNTS = Counter(
    item["read_status"]
    for item in P2AUD_C014_RELEASE_REPORT_EVIDENCE
)

P2AUD_C014_SEARCH_INCOMPLETE = any(
    record["truncated"]
    or record["status"] == "SCANNED_WITH_ERRORS"
    for record in P2AUD_C014_SEARCH_RESULTS
)

P2AUD_C014_REPORT = {
    "schema_version": (
        "P2AUD_FOCUSED_ARTIFACT_AND_DRIFT_REVIEW_V1"
    ),
    "cell_id": "P2AUD-C014",
    "action": 14,
    "status": (
        "P2AUD_C014_FOCUSED_REVIEW_COMPLETE"
        "_WITH_OPEN_FINDINGS"
    ),
    "search_roots": P2AUD_C014_SEARCH_RESULTS,
    "focused_scan_incomplete": (
        P2AUD_C014_SEARCH_INCOMPLETE
    ),
    "missing_paths_rechecked": len(
        P2AUD_C014_TARGET_RECHECK
    ),
    "target_status_counts": dict(
        sorted(P2AUD_C014_TARGET_COUNTS.items())
    ),
    "target_recheck": P2AUD_C014_TARGET_RECHECK,
    "release_reports_inspected": len(
        P2AUD_C014_RELEASE_REPORT_EVIDENCE
    ),
    "release_report_read_counts": dict(
        sorted(P2AUD_C014_REPORT_READ_COUNTS.items())
    ),
    "release_report_evidence": (
        P2AUD_C014_RELEASE_REPORT_EVIDENCE
    ),
    "alternative_contract_test_candidates_count": len(
        P2AUD_C014_TEST_CANDIDATES
    ),
    "alternative_contract_test_examples": (
        P2AUD_C014_TEST_CANDIDATES[:25]
    ),
    "notebook_drift_recheck": (
        P2AUD_C014_DRIFT_RECHECK
    ),
    "previous_variant_drift_count": (
        P2AUD_C013_REPORT["variant_snapshot_drift_count"]
    ),
    "artifact_content_formally_validated": 0,
    "notebooks_formally_accepted": 0,
    "release_formally_accepted": False,
    "formal_gates_passed": 0,
    "files_written": 0,
    "new_api_calls": 0,
    "git_write_operations": 0,
    "next_cell": "P2AUD-C015",
}

assert P2AUD_C014_REPORT["missing_paths_rechecked"] == 6
assert P2AUD_C014_REPORT["release_formally_accepted"] is False

print(
    json.dumps(
        P2AUD_C014_REPORT,
        ensure_ascii=False,
        indent=2,
    )
)
````

````python
# P2AUD-C015 — Build comprehensive Phase 2 As-Built evidence matrix
# Action 15 | READ_ONLY | NO_API | NO_FILE_WRITES | NO_GIT_WRITES

assert P2AUD_C014_REPORT["status"] == (
    "P2AUD_C014_FOCUSED_REVIEW_COMPLETE_WITH_OPEN_FINDINGS"
)
assert len(P2AUD_C011_NOTEBOOK_REGISTRY) == 24
assert len(P2AUD_C013_VARIANT_LEDGER) == 26
assert len(P2AUD_C006_CROSSWALK) == 14

P2AUD_C015_DESCRIPTIONS = {
    "NB00": "محیط و Baseline فاز اول",
    "NB01": "دریافت محتوای وب‌سایت",
    "NB02": "پذیرش و کنترل کیفیت وب‌سایت",
    "NB03": "موجودی کتاب‌ها",
    "NB04": "استخراج و پردازش کتاب‌ها",
    "NB05": "موجودی Transcript",
    "NB06": "پردازش Transcript",
    "NB07": "Schema مشترک و Lineage",
    "NB08": "نرمال‌سازی فارسی",
    "NB09": "حذف تکرار و کیفیت داده",
    "NB10": "Taxonomy و مهندسی دانش",
    "NB11": "آزمایش Chunking",
    "NB12": "Golden Dataset",
    "NB13": "Benchmark مدل Embedding",
    "NB14": "Dense Index",
    "NB15": "Sparse Index",
    "NB16": "Hybrid Retrieval",
    "NB17": "Reranking",
    "NB18": "Context Builder",
    "NB19": "Generation و Citation",
    "NB20": "Answerability، Safety و Confidence",
    "NB21": "Evaluation و Ablation",
    "NB22": "Release و Reproducibility",
    "NB23": "قرارداد تحویل فاز سوم",
}

P2AUD_C015_EXPECTED_IDS = {
    f"NB{i:02d}" for i in range(24)
}

assert set(P2AUD_C015_DESCRIPTIONS) == (
    P2AUD_C015_EXPECTED_IDS
)

# Counts below describe roadmap documentation, not accepted work.

P2AUD_C015_CANDIDATES_BY_NB = Counter(
    candidate["unit_id"]
    for candidate in P2AUD_C004_CANDIDATES
    if candidate["unit_type"] == "NOTEBOOK"
)

P2AUD_C015_ANCHORS_BY_NB = Counter(
    anchor["unit_id"]
    for anchor in P2AUD_C005_ANCHORS
    if anchor["unit_type"] == "NOTEBOOK"
)

P2AUD_C015_PRIMARY_PACKAGES = defaultdict(list)
P2AUD_C015_SUPPORT_PACKAGES = defaultdict(list)

for package in P2AUD_C006_CROSSWALK:
    for notebook_id in package["primary_notebooks"]:
        P2AUD_C015_PRIMARY_PACKAGES[
            notebook_id
        ].append(package["package_id"])

    for notebook_id in package["support_notebooks"]:
        P2AUD_C015_SUPPORT_PACKAGES[
            notebook_id
        ].append(package["package_id"])


P2AUD_C015_VARIANTS_BY_NB = defaultdict(list)
P2AUD_C015_UNCLASSIFIED_VARIANTS = []

for variant in P2AUD_C013_VARIANT_LEDGER:
    notebook_id = variant["family_id"]

    if notebook_id in P2AUD_C015_EXPECTED_IDS:
        P2AUD_C015_VARIANTS_BY_NB[
            notebook_id
        ].append({
            "path": variant["relative_path"],
            "classification": (
                variant["classification_hint"]
            ),
            "snapshot_status": (
                variant["snapshot_status"]
            ),
            "sha256_c011": (
                variant["sha256_from_c011"]
            ),
            "sha256_c013": (
                variant["sha256_current"]
            ),
            "saved_unrun_from_c011": (
                variant["saved_unrun_code_cells"]
            ),
            "saved_errors_from_c011": (
                variant["saved_error_outputs"]
            ),
        })
    else:
        P2AUD_C015_UNCLASSIFIED_VARIANTS.append(
            variant["relative_path"]
        )


P2AUD_C015_FOCUS_BY_NB = defaultdict(list)

for review_id, notebook_ids in (
    P2AUD_C011_FOCUS.items()
):
    for notebook_id in notebook_ids:
        P2AUD_C015_FOCUS_BY_NB[
            notebook_id
        ].append(review_id)


# These are report-location candidates, not proof of successful
# implementation, quality gates, or approved releases.

P2AUD_C015_REPORTS_BY_NB = defaultdict(list)

for report in P2AUD_C014_RELEASE_REPORT_EVIDENCE:
    path = report["relative_path"]
    normalized_path = path.casefold()

    for notebook_id in ("NB22", "NB23"):
        if notebook_id.casefold() in normalized_path:
            P2AUD_C015_REPORTS_BY_NB[
                notebook_id
            ].append({
                "path": path,
                "sha256": report["sha256"],
                "read_status": report["read_status"],
                "evidence_level": (
                    "REPORT_FILE_READ_NOT_GATE_APPROVAL"
                ),
            })


P2AUD_C015_SCHEMA_ARTIFACTS = []

for artifact_id in (
    "RAG_REQUEST_SCHEMA",
    "RAG_RESPONSE_SCHEMA",
):
    record = P2AUD_C012_ARTIFACT_INVENTORY[
        artifact_id
    ]

    P2AUD_C015_SCHEMA_ARTIFACTS.append({
        "artifact_id": artifact_id,
        "path": record["relative_path"],
        "sha256": record["sha256"],
        "file_status": record["status"],
        "contract_runtime_acceptance": "NOT_VERIFIED",
    })


P2AUD_C015_SPECIAL_REMAINING = {
    "NB10": (
        "تعیین مالکیت عملیاتی Taxonomy و اتصال "
        "شواهد پذیرش موجود"
    ),
    "NB20": (
        "بررسی سل‌های خطادار/اجرا‌نشدهٔ نسخهٔ اصلی "
        "در کنار نسخه‌های 20b و 20c"
    ),
    "NB21": (
        "بررسی نسخهٔ اصلی در کنار 21b و 21c؛ "
        "تطبیق گزارش‌های نهایی و تصمیم‌های باز"
    ),
    "NB22": (
        "اتصال آخرین خروجی 22e/22f به NB22؛ "
        "تثبیت Recall@10 و nDCG@10؛ "
        "تعیین تکلیف پذیرش Release"
    ),
    "NB23": (
        "تطبیق قرارداد Draft با Release پذیرفته‌شده؛ "
        "تکمیل شواهد تحویل رسمی فاز سوم"
    ),
}


def p2aud_c015_remaining(notebook_id, flags):
    if notebook_id in P2AUD_C015_SPECIAL_REMAINING:
        return P2AUD_C015_SPECIAL_REMAINING[
            notebook_id
        ]

    if flags:
        return (
            "بررسی پرچم‌های ذخیره‌شده و اتصال "
            "شواهد به Gate مربوط"
        )

    return (
        "تطبیق خروجی‌های اجراشده و گزارش‌های موجود "
        "با الزامات و Gate نقشه"
    )


P2AUD_C015_MATRIX = []

for notebook_id in sorted(P2AUD_C015_EXPECTED_IDS):
    registry = P2AUD_C011_NOTEBOOK_REGISTRY[
        notebook_id
    ]

    execution = P2AUD_C012_NOTEBOOK_EVIDENCE[
        notebook_id
    ]

    assert registry["discovery_status"] == (
        "EXACT_PLANNED_PATH_FOUND"
    )
    assert registry["selected_read_status"] == (
        "READ_SUCCESS"
    )
    assert execution["sha256"] == (
        registry["selected_file_sha256"]
    )

    variants = list(
        P2AUD_C015_VARIANTS_BY_NB[notebook_id]
    )

    main_unrun = execution["saved_unrun_count"]
    main_errors = execution["saved_error_count"]

    flags = []

    if main_unrun:
        flags.append("MAIN_SAVED_UNRUN")

    if main_errors:
        flags.append("MAIN_SAVED_ERRORS")

    if any(
        item["snapshot_status"] == (
            "CHANGED_SINCE_C011"
        )
        for item in variants
    ):
        flags.append("VARIANT_SNAPSHOT_DRIFT")

    if not P2AUD_C015_PRIMARY_PACKAGES[
        notebook_id
    ]:
        flags.append(
            "NO_PRIMARY_OPERATIONAL_PACKAGE"
        )

    if notebook_id == "NB22":
        flags.append(
            "QUALITY_STABILITY_AND_RELEASE_GATE_OPEN"
        )

    if notebook_id == "NB23":
        flags.append(
            "DRAFT_DONE_FORMAL_HANDOFF_OPEN"
        )

    if notebook_id == "NB22":
        delivery_context = (
            "Checkpoint ساخته‌شده؛ "
            "بهبود کیفیت و پذیرش نهایی باز"
        )
    elif notebook_id == "NB23":
        delivery_context = (
            "Draft قرارداد انجام‌شده؛ "
            "تحویل رسمی باز"
        )
    else:
        delivery_context = (
            "پیاده‌سازی در سطح پروژه "
            "انجام‌شده گزارش شده"
        )

    linked_reports = list(
        P2AUD_C015_REPORTS_BY_NB[notebook_id]
    )

    linked_schemas = (
        list(P2AUD_C015_SCHEMA_ARTIFACTS)
        if notebook_id == "NB23"
        else []
    )

    row = {
        "notebook_id": notebook_id,
        "responsibility": (
            P2AUD_C015_DESCRIPTIONS[notebook_id]
        ),
        "roadmap": {
            "planned_path": (
                registry["expected_relative_path"]
            ),
            "source_line": registry["roadmap_line"],
            "source_blob_sha": (
                P2AUD_ROADMAP_INDEX[
                    "PHASE2_MODULAR"
                ]["git_blob_sha"]
            ),
            "requirement_candidates": (
                P2AUD_C015_CANDIDATES_BY_NB[
                    notebook_id
                ]
            ),
            "control_anchors": (
                P2AUD_C015_ANCHORS_BY_NB[
                    notebook_id
                ]
            ),
        },
        "operational_mapping": {
            "primary_packages": sorted(
                P2AUD_C015_PRIMARY_PACKAGES[
                    notebook_id
                ]
            ),
            "support_packages": sorted(
                P2AUD_C015_SUPPORT_PACKAGES[
                    notebook_id
                ]
            ),
            "status": "PROPOSED_NOT_ADJUDICATED",
        },
        "primary_notebook_evidence": {
            "path": registry["selected_path"],
            "sha256_at_c012": execution["sha256"],
            "code_cells": execution["code_cells"],
            "saved_unrun_code_cells": main_unrun,
            "saved_error_outputs": main_errors,
            "static_inspection_status": (
                "FILE_AND_STRUCTURE_VERIFIED_C012"
            ),
            "restart_run_all_verified": False,
        },
        "supplementary_notebooks": variants,
        "supplementary_count": len(variants),
        "candidate_release_reports": linked_reports,
        "candidate_report_count": len(linked_reports),
        "schema_artifacts": linked_schemas,
        "related_review_cases": sorted(
            P2AUD_C015_FOCUS_BY_NB[
                notebook_id
            ]
        ),
        "implementation_context": delivery_context,
        "implementation_status_source": (
            "USER_REPORT_AND_STATIC_INVENTORY"
        ),
        "acceptance_status": (
            "NOT_INDEPENDENTLY_ADJUDICATED"
        ),
        "remaining_audit_work": (
            p2aud_c015_remaining(
                notebook_id, flags
            )
        ),
        "review_flags": flags,
        "notebook_modified_by_this_audit": False,
    }

    P2AUD_C015_MATRIX.append(row)


assert len(P2AUD_C015_MATRIX) == 24
assert {
    row["notebook_id"]
    for row in P2AUD_C015_MATRIX
} == P2AUD_C015_EXPECTED_IDS

P2AUD_C015_ASSIGNED_VARIANTS = sum(
    row["supplementary_count"]
    for row in P2AUD_C015_MATRIX
)

assert (
    P2AUD_C015_ASSIGNED_VARIANTS
    + len(P2AUD_C015_UNCLASSIFIED_VARIANTS)
    == 26
)


# Display a readable 24-row matrix inside the notebook.

def p2aud_c015_escape(value):
    return (
        str(value)
        .replace("|", "\\|")
        .replace("\n", " ")
    )


P2AUD_C015_HEADERS = [
    "NB",
    "مسئولیت",
    "P2 اصلی",
    "کد",
    "اجرا‌نشده",
    "خطا",
    "فرعی",
    "نامزد الزام",
    "شواهد تکمیلی",
    "کار ممیزی باقی‌مانده",
]

P2AUD_C015_TABLE_LINES = [
    "| " + " | ".join(P2AUD_C015_HEADERS) + " |",
    "| " + " | ".join(
        ["---"] * len(P2AUD_C015_HEADERS)
    ) + " |",
]

for row in P2AUD_C015_MATRIX:
    primary_packages = row[
        "operational_mapping"
    ]["primary_packages"]

    package_text = (
        ", ".join(primary_packages)
        if primary_packages
        else "بدون مالک اصلی مستقل"
    )

    evidence_text = []

    if row["candidate_report_count"]:
        evidence_text.append(
            f'{row["candidate_report_count"]} گزارش مرتبط'
        )

    if row["schema_artifacts"]:
        evidence_text.append(
            f'{len(row["schema_artifacts"])} Schema'
        )

    if "VARIANT_SNAPSHOT_DRIFT" in row["review_flags"]:
        evidence_text.append("تغییر نسخهٔ 22e")

    if not evidence_text:
        evidence_text.append(
            "فایل و ساختار بررسی‌شده"
        )

    values = [
        row["notebook_id"],
        row["responsibility"],
        package_text,
        row["primary_notebook_evidence"]["code_cells"],
        row["primary_notebook_evidence"][
            "saved_unrun_code_cells"
        ],
        row["primary_notebook_evidence"][
            "saved_error_outputs"
        ],
        row["supplementary_count"],
        row["roadmap"]["requirement_candidates"],
        "، ".join(evidence_text),
        row["remaining_audit_work"],
    ]

    P2AUD_C015_TABLE_LINES.append(
        "| "
        + " | ".join(
            p2aud_c015_escape(value)
            for value in values
        )
        + " |"
    )

P2AUD_C015_TABLE_MARKDOWN = (
    "### ماتریس جامع NB00 تا NB23\n\n"
    + "\n".join(P2AUD_C015_TABLE_LINES)
    + "\n\n"
    + "**توجه:** این جدول وضعیت مستندات و "
      "شواهد موجود را نشان می‌دهد؛ ستون خطا "
      "و اجرا‌نشده به خروجی ذخیره‌شده مربوط است، "
      "نه الزاماً نتیجهٔ نهایی کار."
)

try:
    from IPython.display import Markdown, display
    display(Markdown(P2AUD_C015_TABLE_MARKDOWN))
except ImportError:
    print(P2AUD_C015_TABLE_MARKDOWN)


P2AUD_C015_FLAG_COUNTS = Counter(
    flag
    for row in P2AUD_C015_MATRIX
    for flag in row["review_flags"]
)

P2AUD_C015_REPORT = {
    "schema_version": (
        "P2AUD_AS_BUILT_MASTER_MATRIX_V1"
    ),
    "cell_id": "P2AUD-C015",
    "action": 15,
    "status": (
        "P2AUD_C015_MASTER_MATRIX_BUILT"
        "_EVIDENCE_REVIEW_PENDING"
    ),
    "planned_notebooks": 24,
    "primary_files_accounted_for": 24,
    "supplementary_notebooks_assigned": (
        P2AUD_C015_ASSIGNED_VARIANTS
    ),
    "supplementary_notebooks_unclassified": (
        P2AUD_C015_UNCLASSIFIED_VARIANTS
    ),
    "roadmap_candidate_count_in_notebook_units": sum(
        row["roadmap"]["requirement_candidates"]
        for row in P2AUD_C015_MATRIX
    ),
    "review_flag_counts": dict(
        sorted(P2AUD_C015_FLAG_COUNTS.items())
    ),
    "main_notebooks_with_saved_execution_flags": [
        row["notebook_id"]
        for row in P2AUD_C015_MATRIX
        if (
            row["primary_notebook_evidence"][
                "saved_unrun_code_cells"
            ] > 0
            or row["primary_notebook_evidence"][
                "saved_error_outputs"
            ] > 0
        )
    ],
    "notebooks_without_primary_operational_package": [
        row["notebook_id"]
        for row in P2AUD_C015_MATRIX
        if not row["operational_mapping"][
            "primary_packages"
        ]
    ],
    "special_status": {
        "NB22": (
            "IMPLEMENTED_CHECKPOINT_QUALITY_STABILITY_OPEN"
        ),
        "NB23": (
            "DRAFT_COMPLETED_FORMAL_HANDOFF_OPEN"
        ),
    },
    "quality_targets": {
        "recall_at_10_target": 0.85,
        "ndcg_at_10_target": 0.75,
        "target_reached_user_report": True,
        "stable_and_formally_accepted": False,
        "latest_run_metric_values_verified": False,
    },
    "source_snapshots": [
        "P2AUD-C002",
        "P2AUD-C006",
        "P2AUD-C011",
        "P2AUD-C012",
        "P2AUD-C013",
        "P2AUD-C014",
    ],
    "row_summary": [
        {
            "notebook": row["notebook_id"],
            "primary_packages": (
                row["operational_mapping"][
                    "primary_packages"
                ]
            ),
            "variants": row["supplementary_count"],
            "saved_unrun": row[
                "primary_notebook_evidence"
            ]["saved_unrun_code_cells"],
            "saved_errors": row[
                "primary_notebook_evidence"
            ]["saved_error_outputs"],
            "review_flags": row["review_flags"],
            "acceptance": row["acceptance_status"],
        }
        for row in P2AUD_C015_MATRIX
    ],
    "implementation_reexecuted": False,
    "formal_gates_newly_accepted": 0,
    "files_written": 0,
    "new_api_calls": 0,
    "git_operations": 0,
    "next_step": (
        "LINK_EXISTING_ACCEPTANCE_AND_ARTIFACT_EVIDENCE"
        "_TO_24_MATRIX_ROWS"
    ),
}

print(
    "\n========== P2AUD-C015 SUMMARY ==========\n"
)
print(
    json.dumps(
        {
            key: value
            for key, value in P2AUD_C015_REPORT.items()
            if key != "row_summary"
        },
        ensure_ascii=False,
        indent=2,
    )
)
````

## اقدام ۱۶ — اتصال شواهد اجرایی به ماتریس As-Built

**شناسه:** `P2AUD-C016`

### هدف
تکمیل ماتریس C015 با ارجاع به گزارش‌های موجود، چک‌لیست‌ها، نسخه‌های تکمیلی و اسناد پذیرش، بدون اجرای دوبارهٔ فعالیت‌های فاز دوم.

### روش
1. حفظ ماتریس C015 بدون تغییر.
2. شناسایی چک‌لیست‌های هر نوت‌بوک.
3. اتصال گزارش‌های دارای شناسهٔ صریح NB به ردیف مربوط.
4. خواندن محدود اسناد کلیدی NB22 برای بررسی بعدی.
5. حفظ تفاوت میان وجود سند، نتیجهٔ تاریخی و پذیرش نهایی.
6. ثبت مواردی که هنوز شواهد کافی ندارند.

### محدودیت
جست‌وجوهای C013 کامل نبوده‌اند؛ بنابراین نبود یک گزارش در فهرست فعلی، اثبات نبود آن در کل پروژه نیست.

### خروجی
- `P2AUD_C016_MATRIX`
- `P2AUD_C016_EVIDENCE_INDEX`
- `P2AUD_C016_PRIORITY_REPORTS`
- `P2AUD_C016_REPORT`

**بدون تغییر فایل، بدون اجرای Pipeline، بدون API و بدون عملیات Git.**

````python
# P2AUD-C016 — Link existing evidence to the As-Built matrix
# Action 16 | READ_ONLY | NO_API | NO_FILE_WRITES | NO_GIT_WRITES
import copy

assert P2AUD_C015_REPORT["status"] == (
    "P2AUD_C015_MASTER_MATRIX_BUILT_EVIDENCE_REVIEW_PENDING"
)
assert len(P2AUD_C015_MATRIX) == 24

P2AUD_C016_ROOT = Path(P2AUD_REPO_ROOT).resolve()
P2AUD_C016_MATRIX = copy.deepcopy(P2AUD_C015_MATRIX)
P2AUD_C016_EVIDENCE_INDEX = {}
P2AUD_C016_PRIORITY_REPORTS = []

P2AUD_C016_NB_PATTERN = re.compile(
    r"(?<![A-Za-z0-9])NB(0[0-9]|1[0-9]|2[0-3])"
    r"(?=[A-Za-z_.\/-]|$)",
    re.IGNORECASE,
)

def p2aud_c016_nb_ids(path):
    return sorted({
        "NB" + match.group(1)
        for match in P2AUD_C016_NB_PATTERN.finditer(path)
    })

# Reuse already collected path inventories.
# Do not repeat the broad filesystem scan.

P2AUD_C016_REPORT_PATHS = sorted({
    path
    for path in (
        list(P2AUD_C013_SCANNED_FILES)
        + list(P2AUD_C014_INDEXED_FILES)
    )
    if path.startswith("reports/phase2/")
    and Path(path).suffix.lower() in {
        ".json", ".md", ".yaml", ".yml", ".txt"
    }
})

P2AUD_C016_REPORTS_BY_NB = defaultdict(list)

for relative_path in P2AUD_C016_REPORT_PATHS:
    for notebook_id in p2aud_c016_nb_ids(relative_path):
        P2AUD_C016_REPORTS_BY_NB[
            notebook_id
        ].append(relative_path)

# Checklists: small, direct directory inventory.

P2AUD_C016_CHECKLISTS_BY_NB = defaultdict(list)
P2AUD_C016_CHECKLIST_DIR = (
    P2AUD_C016_ROOT / "notebooks/phase2/checklist"
)

P2AUD_C016_CHECKLIST_COUNT = 0

if P2AUD_C016_CHECKLIST_DIR.is_dir():
    checklist_paths = sorted(
        path
        for path in P2AUD_C016_CHECKLIST_DIR.iterdir()
        if path.is_file()
        and not path.is_symlink()
        and path.suffix.lower() == ".md"
    )

    if len(checklist_paths) > 500:
        raise ValueError("Checklist directory inventory limit")

    for path in checklist_paths:
        relative_path = path.relative_to(
            P2AUD_C016_ROOT
        ).as_posix()

        notebook_ids = p2aud_c016_nb_ids(path.name)

        if not notebook_ids:
            continue

        P2AUD_C016_CHECKLIST_COUNT += 1

        for notebook_id in notebook_ids:
            P2AUD_C016_CHECKLISTS_BY_NB[
                notebook_id
            ].append(relative_path)


# Read only explicitly identified NB22 priority reports.
# No report status is promoted to approved status.

P2AUD_C016_PRIORITY_PATHS = {
    "NB22_C054_DISPOSITION": (
        "reports/phase2/release/"
        "NB22.C054.quality_disposition/"
        "C054.quality_disposition_supplement."
        "dbf1fd8e5cb2575583351d52e230d5e57f05c7ac61b8a9716e52266d1546d5c2.json"
    ),
    "NB22_C056_METRIC_VERIFICATION": (
        "reports/phase2/release/"
        "NB22.C056.retrieval_verification/"
        "C056.independent_retrieval_metric_verification."
        "ff6d6e136043929e0156e43f6ea75e65bd64f40ab7bcdafef90a4ba40901de22.json"
    ),
    "NB22_C057_ACCEPTANCE_CONTEXT": (
        "reports/phase2/release/"
        "NB22.C057.acceptance_context/"
        "C057.blocked_metric_acceptance_context."
        "3268c0bd3010b50fd8e073cc0b64e1ebcd221a0ad7919dd73dc1d72ee6518995.json"
    ),
    "NB22_C058_REMAINING_WORK": (
        "reports/phase2/release/"
        "NB22.C058.remaining_work/"
        "C058.acceptance_remaining_work_summary."
        "994ac6f05f79fb3feda515c9e00cf0c4084577b81b3cca8eb34fab4cd6880d25.json"
    ),
}

P2AUD_C016_STATUS_KEYS = re.compile(
    r"status|gate|decision|accept|blocked|"
    r"recall|ndcg|metric",
    re.IGNORECASE,
)

def p2aud_c016_extract_fields(value, prefix="", depth=0):
    matches = []

    if not isinstance(value, dict) or depth > 3:
        return matches

    for key, child in value.items():
        field_path = f"{prefix}.{key}" if prefix else str(key)

        if (
            P2AUD_C016_STATUS_KEYS.search(str(key))
            and isinstance(child, (str, int, float, bool))
        ):
            matches.append({
                "field": field_path,
                "value": str(child)[:140],
            })

        if isinstance(child, dict) and depth < 3:
            matches.extend(
                p2aud_c016_extract_fields(
                    child, field_path, depth + 1
                )
            )

        if len(matches) >= 30:
            break

    return matches[:30]


for report_id, relative_path in (
    P2AUD_C016_PRIORITY_PATHS.items()
):
    path = P2AUD_C016_ROOT / relative_path

    record = {
        "report_id": report_id,
        "relative_path": relative_path,
        "read_status": "NOT_FOUND",
        "sha256": None,
        "top_level_keys": [],
        "status_field_examples": [],
        "decision_verified": False,
    }

    try:
        if path.is_symlink():
            record["read_status"] = "SYMLINK_SKIPPED"

        elif path.is_file():
            if path.stat().st_size > 2_000_000:
                record["read_status"] = "SIZE_LIMIT"

            else:
                raw = path.read_bytes()
                content = json.loads(
                    raw.decode("utf-8-sig")
                )

                record["sha256"] = hashlib.sha256(
                    raw
                ).hexdigest()
                record["read_status"] = "READ_SUCCESS"

                if isinstance(content, dict):
                    record["top_level_keys"] = list(
                        content.keys()
                    )[:25]

                    record["status_field_examples"] = (
                        p2aud_c016_extract_fields(
                            content
                        )
                    )

    except (OSError, ValueError, UnicodeError) as exc:
        record["read_status"] = "READ_ERROR"
        record["error_type"] = type(exc).__name__

    P2AUD_C016_PRIORITY_REPORTS.append(record)


# Attach the evidence index to each notebook.
# Preserve original acceptance and implementation judgments.

for row in P2AUD_C016_MATRIX:
    notebook_id = row["notebook_id"]

    checklists = sorted(set(
        P2AUD_C016_CHECKLISTS_BY_NB[
            notebook_id
        ]
    ))

    report_candidates = sorted(set(
        P2AUD_C016_REPORTS_BY_NB[
            notebook_id
        ]
    ))

    priority_reports = [
        record
        for record in P2AUD_C016_PRIORITY_REPORTS
        if notebook_id == "NB22"
    ]

    entry = {
        "notebook_id": notebook_id,
        "primary_path": row[
            "primary_notebook_evidence"
        ]["path"],
        "primary_sha256_at_c012": row[
            "primary_notebook_evidence"
        ]["sha256_at_c012"],
        "supplementary_notebooks": [
            item["path"]
            for item in row["supplementary_notebooks"]
        ],
        "checklist_paths": checklists,
        "explicit_nb_report_candidate_count": len(
            report_candidates
        ),
        "explicit_nb_report_examples": (
            report_candidates[:12]
        ),
        "priority_report_ids": [
            item["report_id"]
            for item in priority_reports
            if item["read_status"] == "READ_SUCCESS"
        ],
        "evidence_index_status": (
            "CANDIDATES_INDEXED_NOT_ADJUDICATED"
        ),
        "implementation_verified_from_reports": False,
        "formal_acceptance_verified": False,
        "coverage_complete": False,
    }

    P2AUD_C016_EVIDENCE_INDEX[
        notebook_id
    ] = entry

    row["evidence_index"] = entry
    row["acceptance_status"] = (
        "NOT_INDEPENDENTLY_ADJUDICATED"
    )


assert len(P2AUD_C016_MATRIX) == 24
assert len(P2AUD_C016_EVIDENCE_INDEX) == 24
assert all(
    not row["evidence_index"]["formal_acceptance_verified"]
    for row in P2AUD_C016_MATRIX
)

P2AUD_C016_REPORT = {
    "schema_version": (
        "P2AUD_AS_BUILT_EVIDENCE_LINKAGE_V1"
    ),
    "cell_id": "P2AUD-C016",
    "status": (
        "P2AUD_C016_EVIDENCE_INDEX_BUILT"
        "_CONTENT_ADJUDICATION_PENDING"
    ),
    "matrix_rows": len(P2AUD_C016_MATRIX),
    "report_paths_examined_from_prior_indexes": len(
        P2AUD_C016_REPORT_PATHS
    ),
    "prior_report_scan_truncated": (
        P2AUD_C013_REPORT["scan_truncated"]
    ),
    "checklists_identified": (
        P2AUD_C016_CHECKLIST_COUNT
    ),
    "priority_reports": [
        {
            "id": item["report_id"],
            "read_status": item["read_status"],
            "sha256": item["sha256"],
            "top_level_keys": item["top_level_keys"],
            "status_field_examples": (
                item["status_field_examples"][:8]
            ),
        }
        for item in P2AUD_C016_PRIORITY_REPORTS
    ],
    "per_notebook": [
        {
            "notebook_id": row["notebook_id"],
            "supplementary_count": (
                row["supplementary_count"]
            ),
            "checklist_count": len(
                row["evidence_index"]["checklist_paths"]
            ),
            "report_candidates": (
                row["evidence_index"][
                    "explicit_nb_report_candidate_count"
                ]
            ),
            "priority_report_ids": (
                row["evidence_index"]["priority_report_ids"]
            ),
            "evidence_status": (
                row["evidence_index"][
                    "evidence_index_status"
                ]
            ),
        }
        for row in P2AUD_C016_MATRIX
    ],
    "reports_formally_adjudicated": 0,
    "notebooks_newly_accepted": 0,
    "gates_newly_passed": 0,
    "files_written": 0,
    "api_calls": 0,
    "git_operations": 0,
    "next_step": (
        "READ_AND_CLASSIFY_EXISTING_EVIDENCE"
        "_WITHOUT_REEXECUTION"
    ),
}

print(
    json.dumps(
        P2AUD_C016_REPORT,
        ensure_ascii=False,
        indent=2,
    )
)
````

````python
# P2AUD-C017 — Classify historical evidence without reexecution
# Action 17 | READ_ONLY | NO_API | NO_FILE_WRITES | NO_GIT_WRITES
import copy
import json
import hashlib
import re
from pathlib import Path
from collections import Counter, defaultdict

assert P2AUD_C016_REPORT["status"] == (
    "P2AUD_C016_EVIDENCE_INDEX_BUILT_CONTENT_ADJUDICATION_PENDING"
)
assert len(P2AUD_C016_MATRIX) == 24

P2AUD_C017_ROOT = Path(P2AUD_REPO_ROOT).resolve()
P2AUD_C017_MATRIX = copy.deepcopy(P2AUD_C016_MATRIX)
P2AUD_C017_DOCUMENT_LEDGER = []
P2AUD_C017_SOURCE_JSON = {}
P2AUD_C017_CHECKLIST_LEDGER = []

# Select a small, explicitly known set of historical reports.
# No broad report search or notebook execution.

P2AUD_C017_DOC_SPECS = []

for record in P2AUD_C016_PRIORITY_REPORTS:
    P2AUD_C017_DOC_SPECS.append({
        "doc_id": record["report_id"],
        "notebook_id": "NB22",
        "relative_path": record["relative_path"],
        "baseline_sha256": record["sha256"],
        "source_cell": "C016",
    })

P2AUD_C017_EXTRA_PATTERNS = {
    "NB22_C047_FINAL_GATE": (
        "NB22", "C047.final_release_gate."
    ),
    "NB22_C049_HANDOFF_RESULT": (
        "NB22", "C049.handoff_result."
    ),
    "NB23_C012_DRAFT_READINESS": (
        "NB23", "NB23.C012.normalized_draft_readiness."
    ),
}

for doc_id, (notebook_id, fragment) in (
    P2AUD_C017_EXTRA_PATTERNS.items()
):
    matches = [
        record
        for record in P2AUD_C014_RELEASE_REPORT_EVIDENCE
        if fragment in record["relative_path"]
    ]

    if len(matches) == 1:
        record = matches[0]

        P2AUD_C017_DOC_SPECS.append({
            "doc_id": doc_id,
            "notebook_id": notebook_id,
            "relative_path": record["relative_path"],
            "baseline_sha256": record["sha256"],
            "source_cell": "C014",
        })

    else:
        P2AUD_C017_DOCUMENT_LEDGER.append({
            "doc_id": doc_id,
            "notebook_id": notebook_id,
            "read_status": "SOURCE_NOT_UNIQUELY_IDENTIFIED",
            "matching_sources": len(matches),
            "interpretation": "NOT_VERIFIED",
        })


def p2aud_c017_get_field(data, dotted_path):
    current = data

    for key in dotted_path.split("."):
        if not isinstance(current, dict):
            return None

        if key not in current:
            return None

        current = current[key]

    return current


P2AUD_C017_FIELDS = (
    "status",
    "release_acceptance_granted",
    "phase2_completion_claimed",
    "all_blocked_metrics_resolved",
    "blocked_metric_count",
    "new_performance_improvement_claimed",
    "independent_recomputation_complete",
    "all_recorded_metric_comparisons_match",
    "release_gate_modified",
    "computed_metrics.test.recall_at_10",
    "computed_metrics.test.ndcg_at_10",
    "contract.configuration.retrieval_recall_at_10_min",
    "contract.configuration.retrieval_ndcg_at_10_min",
    "metric_status_accounting.recorded_gate_counts",
)

for spec in P2AUD_C017_DOC_SPECS:
    relative_path = spec["relative_path"]
    path = P2AUD_C017_ROOT / relative_path

    result = {
        **spec,
        "read_status": "NOT_READ",
        "current_sha256": None,
        "snapshot_status": "NOT_CHECKED",
        "extracted_fields": {},
        "interpretation": (
            "HISTORICAL_DOCUMENT_NOT_CURRENT_RELEASE_APPROVAL"
        ),
    }

    try:
        if path.is_symlink():
            result["read_status"] = "SYMLINK_SKIPPED"

        elif not path.is_file():
            result["read_status"] = "NOT_FOUND"

        elif path.stat().st_size > 2_000_000:
            result["read_status"] = "SIZE_LIMIT"

        else:
            raw = path.read_bytes()
            current_sha = hashlib.sha256(raw).hexdigest()

            result["current_sha256"] = current_sha
            result["snapshot_status"] = (
                "MATCH_BASELINE"
                if current_sha == spec["baseline_sha256"]
                else "CHANGED_SINCE_BASELINE"
            )

            if path.suffix.lower() == ".json":
                data = json.loads(raw.decode("utf-8-sig"))

                if not isinstance(data, dict):
                    raise ValueError("Expected JSON object")

                P2AUD_C017_SOURCE_JSON[
                    spec["doc_id"]
                ] = data

                extracted = {}

                for field in P2AUD_C017_FIELDS:
                    value = p2aud_c017_get_field(
                        data, field
                    )

                    if value is not None:
                        extracted[field] = value

                result["extracted_fields"] = extracted
                result["read_status"] = "READ_SUCCESS"

            elif path.suffix.lower() == ".md":
                content = raw.decode("utf-8-sig")

                result["read_status"] = "READ_SUCCESS"
                result["markdown_lines"] = len(
                    content.splitlines()
                )
                result["draft_word_detected"] = bool(
                    re.search(
                        r"\bdraft\b|پیش‌نویس",
                        content,
                        re.IGNORECASE,
                    )
                )

            else:
                result["read_status"] = (
                    "UNSUPPORTED_FILE_TYPE"
                )

    except (OSError, ValueError, UnicodeError) as exc:
        result["read_status"] = "READ_ERROR"
        result["error_type"] = type(exc).__name__

    P2AUD_C017_DOCUMENT_LEDGER.append(result)


# Read small known checklists, without interpreting checkmarks
# as product-level acceptance.

P2AUD_C017_CHECKLIST_PATHS = sorted({
    path
    for paths in P2AUD_C016_CHECKLISTS_BY_NB.values()
    for path in paths
})

for relative_path in P2AUD_C017_CHECKLIST_PATHS:
    path = P2AUD_C017_ROOT / relative_path

    record = {
        "relative_path": relative_path,
        "read_status": "NOT_READ",
        "sha256": None,
        "headings": [],
        "checked_checkbox_count": None,
        "open_checkbox_count": None,
        "document_acceptance": "NOT_ADJUDICATED",
    }

    try:
        if path.is_symlink():
            record["read_status"] = "SYMLINK_SKIPPED"

        elif not path.is_file():
            record["read_status"] = "NOT_FOUND"

        elif path.stat().st_size > 1_000_000:
            record["read_status"] = "SIZE_LIMIT"

        else:
            raw = path.read_bytes()
            content = raw.decode("utf-8-sig")

            record["sha256"] = hashlib.sha256(
                raw
            ).hexdigest()
            record["read_status"] = "READ_SUCCESS"

            record["checked_checkbox_count"] = len(
                re.findall(
                    r"(?m)^\s*-\s*\[[xX]\]",
                    content
                )
            )

            record["open_checkbox_count"] = len(
                re.findall(
                    r"(?m)^\s*-\s*\[\s\]",
                    content
                )
            )

            headings = [
                line.strip()[:160]
                for line in content.splitlines()
                if re.match(r"^#{1,3}\s+", line)
            ]

            record["headings"] = headings[:8]

    except (OSError, UnicodeError) as exc:
        record["read_status"] = "READ_ERROR"
        record["error_type"] = type(exc).__name__

    P2AUD_C017_CHECKLIST_LEDGER.append(record)


# Preserve historical metric values with explicit provenance.
# Do not substitute newer unverified experimental results.

P2AUD_C017_C056 = P2AUD_C017_SOURCE_JSON.get(
    "NB22_C056_METRIC_VERIFICATION", {}
)

P2AUD_C017_C057 = P2AUD_C017_SOURCE_JSON.get(
    "NB22_C057_ACCEPTANCE_CONTEXT", {}
)

P2AUD_C017_C058 = P2AUD_C017_SOURCE_JSON.get(
    "NB22_C058_REMAINING_WORK", {}
)

P2AUD_C017_C054 = P2AUD_C017_SOURCE_JSON.get(
    "NB22_C054_DISPOSITION", {}
)

historical_test = (
    P2AUD_C017_C056.get("computed_metrics", {})
    .get("test", {})
)

frozen_config = (
    P2AUD_C017_C057.get("contract", {})
    .get("configuration", {})
)

gate_counts = (
    P2AUD_C017_C058.get("metric_status_accounting", {})
    .get("recorded_gate_counts", {})
)

P2AUD_C017_HISTORICAL_QUALITY = {
    "metric_source": "NB22_C056_METRIC_VERIFICATION",
    "contract_source": "NB22_C057_ACCEPTANCE_CONTEXT",
    "gate_source": "NB22_C058_REMAINING_WORK",
    "historical_test_recall_at_10": (
        historical_test.get("recall_at_10")
    ),
    "historical_test_ndcg_at_10": (
        historical_test.get("ndcg_at_10")
    ),
    "frozen_recall_at_10_min": (
        frozen_config.get("retrieval_recall_at_10_min")
    ),
    "frozen_ndcg_at_10_min": (
        frozen_config.get("retrieval_ndcg_at_10_min")
    ),
    "recorded_gate_counts": gate_counts,
    "recorded_blocked_metric_count": (
        P2AUD_C017_C057.get("blocked_metric_count")
    ),
    "historical_release_acceptance_c054": (
        P2AUD_C017_C054.get("release_acceptance_granted")
    ),
    "historical_release_acceptance_c058": (
        P2AUD_C017_C058.get("release_acceptance_granted")
    ),
    "newer_targets_reached": (
        "REPORTED_BY_PROJECT_OWNER_NOT_VERSION_VERIFIED"
    ),
    "newer_result_stability": "NOT_VERIFIED",
    "newer_result_holdout_acceptance": "NOT_VERIFIED",
}

recall_value = P2AUD_C017_HISTORICAL_QUALITY[
    "historical_test_recall_at_10"
]
recall_min = P2AUD_C017_HISTORICAL_QUALITY[
    "frozen_recall_at_10_min"
]
ndcg_value = P2AUD_C017_HISTORICAL_QUALITY[
    "historical_test_ndcg_at_10"
]
ndcg_min = P2AUD_C017_HISTORICAL_QUALITY[
    "frozen_ndcg_at_10_min"
]

P2AUD_C017_HISTORICAL_QUALITY[
    "historical_recall_threshold_pass"
] = (
    recall_value >= recall_min
    if isinstance(recall_value, (int, float))
    and isinstance(recall_min, (int, float))
    else None
)

P2AUD_C017_HISTORICAL_QUALITY[
    "historical_ndcg_threshold_pass"
] = (
    ndcg_value >= ndcg_min
    if isinstance(ndcg_value, (int, float))
    and isinstance(ndcg_min, (int, float))
    else None
)


# Enrich the existing matrix, preserving original rows.
# Do not infer formal notebook acceptance from file existence.

P2AUD_C017_DOCS_BY_NB = defaultdict(list)

for document in P2AUD_C017_DOCUMENT_LEDGER:
    if document.get("read_status") == "READ_SUCCESS":
        P2AUD_C017_DOCS_BY_NB[
            document["notebook_id"]
        ].append(document["doc_id"])


P2AUD_C017_STATUS_DESCRIPTIONS = {
    "NB20": (
        "پیاده‌سازی گزارش‌شده؛ "
        "پرچم‌های ذخیره‌شدهٔ نسخهٔ اصلی "
        "نیازمند تطبیق با 20b/20c"
    ),
    "NB21": (
        "ارزیابی و اقدامات تاریخی انجام‌شده؛ "
        "پذیرش کیفیت از تکمیل فعالیت جداست"
    ),
    "NB22": (
        "Checkpoint تولیدشده؛ "
        "Gate تاریخی کیفیت باز؛ "
        "نتیجهٔ جدید نیازمند اتصال نسخه و ارزیابی پایداری"
    ),
    "NB23": (
        "Draft قرارداد آماده؛ "
        "تحویل رسمی وابسته به Release پذیرفته‌شده"
    ),
}

for row in P2AUD_C017_MATRIX:
    notebook_id = row["notebook_id"]

    checklist_paths = row[
        "evidence_index"
    ]["checklist_paths"]

    row["as_built_evidence_status"] = {
        "implementation_reported_by_owner": True,
        "primary_file_structure_verified": True,
        "historical_documents_reviewed": (
            notebook_id in {
                "NB20", "NB21", "NB22", "NB23"
            }
        ),
        "linked_checklist_count": len(checklist_paths),
        "key_report_ids_read": (
            P2AUD_C017_DOCS_BY_NB[notebook_id]
        ),
        "historical_work_description": (
            P2AUD_C017_STATUS_DESCRIPTIONS.get(
                notebook_id,
                "پیاده‌سازی در سطح پروژه گزارش شده؛ "
                "شواهد پذیرش این ردیف هنوز بررسی تفصیلی نشده"
            )
        ),
        "latest_execution_formally_verified": False,
        "formal_acceptance": "NOT_VERIFIED_CURRENT",
        "historical_flags_are_current_failures": (
            "NOT_DETERMINED"
        ),
        "classification_basis": (
            "OWNER_REPORT_PLUS_VERSIONED_HISTORICAL_EVIDENCE"
            if notebook_id in {"NB20", "NB21", "NB22", "NB23"}
            else "OWNER_REPORT_PLUS_STATIC_NOTEBOOK_INVENTORY"
        ),
    }

    if notebook_id == "NB22":
        row["as_built_evidence_status"][
            "historical_quality"
        ] = P2AUD_C017_HISTORICAL_QUALITY


assert len(P2AUD_C017_MATRIX) == 24

P2AUD_C017_DOCUMENT_STATUS_COUNTS = Counter(
    record["read_status"]
    for record in P2AUD_C017_DOCUMENT_LEDGER
)

P2AUD_C017_CHECKLIST_STATUS_COUNTS = Counter(
    record["read_status"]
    for record in P2AUD_C017_CHECKLIST_LEDGER
)

P2AUD_C017_REPORT = {
    "schema_version": (
        "P2AUD_AS_BUILT_HISTORICAL_EVIDENCE_CLASSIFICATION_V1"
    ),
    "cell_id": "P2AUD-C017",
    "action": 17,
    "status": (
        "P2AUD_C017_HISTORICAL_CONTEXT_ATTACHED"
        "_LATEST_RELEASE_VALIDATION_OPEN"
    ),
    "matrix_rows_preserved": len(P2AUD_C017_MATRIX),
    "targeted_documents": len(
        P2AUD_C017_DOCUMENT_LEDGER
    ),
    "document_status_counts": dict(
        sorted(P2AUD_C017_DOCUMENT_STATUS_COUNTS.items())
    ),
    "checklists_reviewed": len(
        P2AUD_C017_CHECKLIST_LEDGER
    ),
    "checklist_status_counts": dict(
        sorted(P2AUD_C017_CHECKLIST_STATUS_COUNTS.items())
    ),
    "historical_quality": (
        P2AUD_C017_HISTORICAL_QUALITY
    ),
    "priority_notebook_summary": [
        {
            "notebook_id": row["notebook_id"],
            "status": row[
                "as_built_evidence_status"
            ]["historical_work_description"],
            "key_report_ids": row[
                "as_built_evidence_status"
            ]["key_report_ids_read"],
            "formal_acceptance": row[
                "as_built_evidence_status"
            ]["formal_acceptance"],
        }
        for row in P2AUD_C017_MATRIX
        if row["notebook_id"] in {
            "NB20", "NB21", "NB22", "NB23"
        }
    ],
    "open_roadmap_difference_cases": 5,
    "current_22e_snapshot": {
        "sha256": (
            P2AUD_C014_DRIFT_RECHECK["sha256_c014"]
        ),
        "code_cells": (
            P2AUD_C014_DRIFT_RECHECK["code_cells_c014"]
        ),
        "saved_unrun": (
            P2AUD_C014_DRIFT_RECHECK["unrun_cells_c014"]
        ),
        "saved_errors": (
            P2AUD_C014_DRIFT_RECHECK["error_outputs_c014"]
        ),
    },
    "latest_22e_22f_results_accepted": False,
    "formal_gates_newly_passed": 0,
    "files_written": 0,
    "new_api_calls": 0,
    "git_operations": 0,
    "next_step": (
        "TRACE_LATEST_22E_22F_METRICS_TO_VERSIONED_OUTPUTS"
    ),
}

print(
    json.dumps(
        P2AUD_C017_REPORT,
        ensure_ascii=False,
        indent=2,
    )
)
````

## P2AUD-C018 — ردیابی نتایج جدید 22e و 22f

### هدف
اتصال شواهد آخرین آزمایش‌های بازیابی به ماتریس As-Built، بدون بازاجرای آزمایش.

### کنترل‌ها
1. خواندن دو نوت‌بوک 22e-v2 و 22f-v1.
2. ثبت SHA-256 و تغییر احتمالی نسخه.
3. تفکیک اعداد موجود در کد از اعداد موجود در خروجی ذخیره‌شده.
4. استخراج Recall@10 و nDCG@10 با نشانی دقیق سل و خروجی.
5. ثبت نوع دادهٔ ارزیابی، در صورت وجود نشانهٔ صریح.
6. ثبت Error Output و سل‌های اجرا‌نشده.
7. اتصال نتایج به ردیف NB22 در ماتریس.

### قاعدهٔ پذیرش
وجود عدد 0.85 یا 0.75، حتی در خروجی ذخیره‌شده، به‌تنهایی اثبات PASS نیست. نسخهٔ داده، Split، تعریف سنجه و پایداری نتیجه باید مشخص باشند.

**بدون API، اجرای مدل، تغییر فایل یا Git.**

</div>

````python
# P2AUD-C018 — Trace NB22e/NB22f saved metrics and provenance
# READ_ONLY | NO_API | NO_FILE_WRITES | NO_GIT_WRITES

import json
import re
import hashlib
import copy
from pathlib import Path
from collections import Counter

assert P2AUD_C017_REPORT["status"] == (
    "P2AUD_C017_HISTORICAL_CONTEXT_ATTACHED_LATEST_RELEASE_VALIDATION_OPEN"
)
assert len(P2AUD_C017_MATRIX) == 24

P2AUD_C018_ROOT = Path(P2AUD_REPO_ROOT).resolve()
P2AUD_C018_MAX_FILE_BYTES = 64 * 1024 * 1024
P2AUD_C018_MAX_OUTPUT_CHARS = 2_000_000

P2AUD_C018_TARGETS = {
    "NB22E_V2": {
        "path": (
            "notebooks/phase2/"
            "22e_retrieval_experiment_lab_v2.ipynb"
        ),
        "baseline_sha256": (
            P2AUD_C014_DRIFT_RECHECK["sha256_c014"]
        ),
        "baseline_cell": "C014",
    },
    "NB22F_V1": {
        "path": (
            "notebooks/phase2/"
            "22f_consolidated_retrieval_benchmark_v1.ipynb"
        ),
        "baseline_sha256": next(
            item["sha256"]
            for item in P2AUD_C011_VARIANT_REGISTRY
            if item["relative_path"].endswith(
                "/22f_consolidated_retrieval_benchmark_v1.ipynb"
            )
        ),
        "baseline_cell": "C011",
    },
}

P2AUD_C018_DIGITS = str.maketrans(
    "۰۱۲۳۴۵۶۷۸۹٠١٢٣٤٥٦٧٨٩٪٫",
    "01234567890123456789%."
)

P2AUD_C018_METRIC_REGEX = re.compile(
    r"(?P<metric>"
    r"recall(?:\s*@\s*10|[_\s-]*at[_\s-]*10)"
    r"|ndcg(?:\s*@\s*10|[_\s-]*at[_\s-]*10)"
    r")"
    r"\s*[\"']?\s*[:=]\s*"
    r"[\"']?(?P<value>\d+(?:\.\d+)?)"
    r"\s*(?P<percent>%?)",
    re.IGNORECASE,
)

P2AUD_C018_SPLIT_REGEX = re.compile(
    r"\b(holdout|development|validation|test|dev|train)\b",
    re.IGNORECASE,
)

P2AUD_C018_TARGET_REGEX = re.compile(
    r"\b(target|threshold|minimum|baseline|"
    r"acceptance|min_recall|min_ndcg)\b",
    re.IGNORECASE,
)


def p2aud_c018_text(value):
    if isinstance(value, str):
        return value

    if isinstance(value, list):
        return "".join(
            part for part in value
            if isinstance(part, str)
        )

    return ""


def p2aud_c018_output_texts(output):
    if not isinstance(output, dict):
        return []

    result = []

    if output.get("output_type") == "stream":
        result.append((
            "STREAM",
            p2aud_c018_text(output.get("text", ""))
        ))

    data = output.get("data", {})

    if isinstance(data, dict):
        for mime_type in (
            "text/plain",
            "application/json",
            "text/markdown",
            "text/html",
        ):
            if mime_type not in data:
                continue

            value = data[mime_type]

            if isinstance(value, (str, list)):
                result.append((
                    mime_type,
                    p2aud_c018_text(value)
                ))
            elif isinstance(value, dict):
                result.append((
                    mime_type,
                    json.dumps(value, ensure_ascii=False)
                ))

    return result


def p2aud_c018_metric_hits(
    text,
    cell_number,
    origin,
    output_index=None,
    execution_count=None,
):
    normalized = text.translate(P2AUD_C018_DIGITS)
    results = []

    for match in P2AUD_C018_METRIC_REGEX.finditer(
        normalized
    ):
        metric_name = (
            "RECALL_AT_10"
            if match.group("metric").lower().startswith("recall")
            else "NDCG_AT_10"
        )

        value = float(match.group("value"))

        context_start = max(0, match.start() - 100)
        context_end = min(
            len(normalized),
            match.end() + 100
        )

        context = normalized[
            context_start:context_end
        ]

        split_hints = sorted(set(
            token.lower()
            for token in P2AUD_C018_SPLIT_REGEX.findall(
                context
            )
        ))

        results.append({
            "metric": metric_name,
            "observed_value": value,
            "percent_symbol_present": bool(
                match.group("percent")
            ),
            "cell_number": cell_number,
            "origin": origin,
            "output_index": output_index,
            "saved_execution_count": execution_count,
            "split_hints_near_value": split_hints,
            "threshold_or_baseline_hint": bool(
                P2AUD_C018_TARGET_REGEX.search(context)
            ),
            "value_role": "NOT_YET_ADJUDICATED",
            "release_acceptance_proof": False,
        })

    return results


P2AUD_C018_NOTEBOOK_EVIDENCE = {}
P2AUD_C018_ALL_METRIC_HITS = []

for notebook_key, spec in P2AUD_C018_TARGETS.items():
    relative_path = spec["path"]
    path = P2AUD_C018_ROOT / relative_path

    record = {
        "notebook_key": notebook_key,
        "path": relative_path,
        "read_status": "NOT_READ",
        "baseline_sha256": spec["baseline_sha256"],
        "baseline_cell": spec["baseline_cell"],
        "current_sha256": None,
        "snapshot_status": "NOT_CHECKED",
        "code_cells": None,
        "markdown_cells": None,
        "saved_unrun_code_cells": None,
        "saved_error_outputs": None,
        "code_metric_hit_count": 0,
        "output_metric_hit_count": 0,
        "code_metric_examples": [],
        "saved_output_metric_examples": [],
        "saved_output_metric_latest_examples": [],
        "skipped_oversized_outputs": 0,
        "execution_reperformed": False,
        "quality_acceptance_verified": False,
    }

    if path.is_symlink():
        record["read_status"] = "SYMLINK_SKIPPED"
        P2AUD_C018_NOTEBOOK_EVIDENCE[
            notebook_key
        ] = record
        continue

    if not path.is_file():
        record["read_status"] = "FILE_NOT_FOUND"
        P2AUD_C018_NOTEBOOK_EVIDENCE[
            notebook_key
        ] = record
        continue

    assert path.resolve().is_relative_to(
        P2AUD_C018_ROOT
    )

    try:
        if path.stat().st_size > P2AUD_C018_MAX_FILE_BYTES:
            record["read_status"] = "SIZE_LIMIT"
            P2AUD_C018_NOTEBOOK_EVIDENCE[
                notebook_key
            ] = record
            continue

        raw = path.read_bytes()
        current_sha = hashlib.sha256(raw).hexdigest()

        record["current_sha256"] = current_sha
        record["snapshot_status"] = (
            "MATCH_" + spec["baseline_cell"]
            if current_sha == spec["baseline_sha256"]
            else "CHANGED_SINCE_" + spec["baseline_cell"]
        )

        notebook = json.loads(
            raw.decode("utf-8-sig")
        )

        cells = notebook.get("cells")

        if not isinstance(cells, list):
            raise ValueError("Invalid notebook cells")

        counts = Counter()
        code_hits = []
        output_hits = []

        for cell_number, cell in enumerate(
            cells, start=1
        ):
            if not isinstance(cell, dict):
                continue

            cell_type = cell.get("cell_type")

            if cell_type == "markdown":
                counts["markdown"] += 1
                continue

            if cell_type != "code":
                continue

            counts["code"] += 1

            source = p2aud_c018_text(
                cell.get("source", "")
            )

            execution_count = cell.get(
                "execution_count"
            )

            if source.strip() and execution_count is None:
                counts["unrun"] += 1

            if len(source) <= P2AUD_C018_MAX_OUTPUT_CHARS:
                code_hits.extend(
                    p2aud_c018_metric_hits(
                        source,
                        cell_number,
                        "CODE_SOURCE",
                        execution_count=execution_count,
                    )
                )

            outputs = cell.get("outputs", [])

            if not isinstance(outputs, list):
                continue

            for output_index, output in enumerate(
                outputs, start=1
            ):
                if not isinstance(output, dict):
                    continue

                if output.get("output_type") == "error":
                    counts["errors"] += 1

                for mime_type, output_text in (
                    p2aud_c018_output_texts(output)
                ):
                    if (
                        len(output_text)
                        > P2AUD_C018_MAX_OUTPUT_CHARS
                    ):
                        record[
                            "skipped_oversized_outputs"
                        ] += 1
                        continue

                    output_hits.extend(
                        p2aud_c018_metric_hits(
                            output_text,
                            cell_number,
                            "SAVED_OUTPUT_" + mime_type,
                            output_index=output_index,
                            execution_count=execution_count,
                        )
                    )

        record.update({
            "read_status": "READ_SUCCESS",
            "code_cells": counts["code"],
            "markdown_cells": counts["markdown"],
            "saved_unrun_code_cells": counts["unrun"],
            "saved_error_outputs": counts["errors"],
            "code_metric_hit_count": len(code_hits),
            "output_metric_hit_count": len(output_hits),
            "code_metric_examples": code_hits[:8],
            "saved_output_metric_examples": output_hits[:10],
            "saved_output_metric_latest_examples": (
                output_hits[-10:]
            ),
        })

        P2AUD_C018_ALL_METRIC_HITS.extend([
            {
                "notebook_key": notebook_key,
                "notebook_sha256": current_sha,
                **hit,
            }
            for hit in output_hits
        ])

    except (
        OSError, ValueError, TypeError, UnicodeError
    ) as exc:
        record["read_status"] = "READ_ERROR"
        record["error_type"] = type(exc).__name__

    P2AUD_C018_NOTEBOOK_EVIDENCE[
        notebook_key
    ] = record


# Attach to NB22 without altering historical metrics.

P2AUD_C018_MATRIX = copy.deepcopy(
    P2AUD_C017_MATRIX
)

P2AUD_C018_NB22 = next(
    row
    for row in P2AUD_C018_MATRIX
    if row["notebook_id"] == "NB22"
)

P2AUD_C018_NB22["latest_experiment_evidence"] = {
    "notebooks": P2AUD_C018_NOTEBOOK_EVIDENCE,
    "historical_test_metrics_preserved": True,
    "new_metric_candidates_found": len(
        P2AUD_C018_ALL_METRIC_HITS
    ),
    "metric_values_adjudicated": 0,
    "dataset_split_verified": False,
    "model_version_verified": False,
    "stable_reproduction_verified": False,
    "independent_holdout_verified": False,
    "formal_release_acceptance": "NOT_VERIFIED",
}

P2AUD_C018_READ_COUNTS = Counter(
    record["read_status"]
    for record in P2AUD_C018_NOTEBOOK_EVIDENCE.values()
)

P2AUD_C018_REPORT = {
    "schema_version": (
        "P2AUD_NB22_LATEST_METRIC_PROVENANCE_V1"
    ),
    "cell_id": "P2AUD-C018",
    "action": 18,
    "status": (
        "P2AUD_C018_SAVED_EVIDENCE_INSPECTED"
        "_METRIC_PROVENANCE_REVIEW_PENDING"
    ),
    "notebooks_targeted": 2,
    "read_status_counts": dict(
        sorted(P2AUD_C018_READ_COUNTS.items())
    ),
    "notebook_summary": {
        key: record
        for key, record in (
            P2AUD_C018_NOTEBOOK_EVIDENCE.items()
        )
    },
    "saved_output_metric_candidates": len(
        P2AUD_C018_ALL_METRIC_HITS
    ),
    "historical_test_recall_at_10": (
        P2AUD_C017_HISTORICAL_QUALITY[
            "historical_test_recall_at_10"
        ]
    ),
    "historical_test_ndcg_at_10": (
        P2AUD_C017_HISTORICAL_QUALITY[
            "historical_test_ndcg_at_10"
        ]
    ),
    "recall_at_10_target": 0.85,
    "ndcg_at_10_target": 0.75,
    "latest_scores_independently_verified": False,
    "latest_release_accepted": False,
    "matrix_rows_preserved": len(
        P2AUD_C018_MATRIX
    ),
    "files_written": 0,
    "api_calls": 0,
    "git_operations": 0,
    "next_step": (
        "IDENTIFY_VERSIONED_RESULT_ARTIFACTS"
        "_AND_EXACT_EVALUATION_SPLIT"
    ),
}

assert len(P2AUD_C018_MATRIX) == 24

print(
    json.dumps(
        P2AUD_C018_REPORT,
        ensure_ascii=False,
        indent=2,
    )
)
````

## P2AUD-C019 — تعیین منشأ نتایج 22e/22f

### هدف
تشخیص تفاوت میان نتایج موردی، خلاصه‌های آزمایشی و شواهد قابل استفاده برای پذیرش کیفیت.

2. ثبت خط، سل، Output و هش هر شاهد عددی.
3. استخراج نشانه‌های Split و Summary از متن مجاور.
4. شناسایی ارجاعات به فایل‌های نتیجه و Manifest.
5. تطبیق ارجاعات با فایل‌های موجود، بدون جست‌وجوی گسترده.
6. اتصال وضعیت شواهد به ردیف NB22 ماتریس.

### محدودیت
هیچ عددی صرفاً به دلیل بزرگ‌تر بودن از آستانه، نتیجهٔ نهایی یا PASS تلقی نمی‌شود.

**READ_ONLY — NO_API — NO_FILE_WRITES — NO_GIT_WRITES**

</div>

````python
# P2AUD-C019 — Trace metric contexts, splits and result artifacts
# READ_ONLY | NO_API | NO_FILE_WRITES | NO_GIT_WRITES

import json
import re
import hashlib
import copy
from pathlib import Path
from collections import Counter, defaultdict

assert P2AUD_C018_REPORT["status"] == (
    "P2AUD_C018_SAVED_EVIDENCE_INSPECTED"
    "_METRIC_PROVENANCE_REVIEW_PENDING"
)
assert len(P2AUD_C018_MATRIX) == 24

P2AUD_C019_ROOT = Path(P2AUD_REPO_ROOT).resolve()
P2AUD_C019_MAX_BYTES = 64 * 1024 * 1024
P2AUD_C019_MAX_OUTPUT_CHARS = 2_000_000

P2AUD_C019_METRIC_CONTEXT = []
P2AUD_C019_ARTIFACT_REFERENCES = []
P2AUD_C019_NOTEBOOK_STATUS = {}
P2AUD_C019_OUTPUT_GROUPS = defaultdict(list)

P2AUD_C019_SPLIT_RE = re.compile(
    r"\b(?:holdout|development|validation|test|dev|train)\b",
    re.I,
)

P2AUD_C019_SUMMARY_RE = re.compile(
    r"\b(?:summary|aggregate|aggregated|macro|"
    r"overall|final|selected|best|mean)\b",
    re.I,
)

P2AUD_C019_PATH_RE = re.compile(
    r"(?<![\w./])"
    r"(?:reports|artifacts|data|schemas|tests|config)"
    r"/[A-Za-z0-9_./{}-]{3,220}"
    r"\.(?:jsonl|json|csv|parquet|md|yaml|yml|npz)\b",
    re.I,
)


def p2aud_c019_extract_paths(text, source, notebook_id):
    for match in P2AUD_C019_PATH_RE.finditer(text):
        relative_path = match.group(0).rstrip(".,;:")

        if ".." in Path(relative_path).parts:
            continue

        P2AUD_C019_ARTIFACT_REFERENCES.append({
            "notebook": notebook_id,
            "path_literal": relative_path,
            "origin": source,
            "reference_only": True,
        })


def p2aud_c019_extract_output_metrics(
    text,
    notebook_id,
    file_sha,
    cell_number,
    output_number,
    output_kind,
    execution_count,
):
    normalized = text.translate(P2AUD_C018_DIGITS)
    lines = normalized.splitlines()

    for index, line in enumerate(lines):
        matches = list(
            P2AUD_C018_METRIC_REGEX.finditer(line)
        )

        if not matches:
            continue

        # Nearby labels are clues, not verified population lineage.
        preceding = lines[max(0, index - 4):index]
        context = "\n".join(preceding + [line])

        split_hints = sorted(set(
            match.lower()
            for match in P2AUD_C019_SPLIT_RE.findall(
                context
            )
        ))

        summary_hint = bool(
            P2AUD_C019_SUMMARY_RE.search(context)
        )

        for match in matches:
            metric = (
                "RECALL_AT_10"
                if match.group("metric").lower().startswith(
                    "recall"
                )
                else "NDCG_AT_10"
            )

            value = float(match.group("value"))

            # Preserve only bounded context near the metric.
            start = max(0, match.start() - 110)
            end = min(len(line), match.end() + 110)

            snippet = line[start:end].strip()

            evidence = {
                "notebook": notebook_id,
                "file_sha256": file_sha,
                "cell_number": cell_number,
                "output_number": output_number,
                "output_kind": output_kind,
                "execution_count": execution_count,
                "output_line": index + 1,
                "metric": metric,
                "value": value,
                "context_excerpt": snippet[:240],
                "nearby_split_hints": split_hints,
                "summary_word_nearby": summary_hint,
                "population_verified": False,
                "aggregate_verified": False,
                "selected_candidate_verified": False,
                "formal_gate_pass": False,
            }

            P2AUD_C019_METRIC_CONTEXT.append(evidence)

            group_key = (
                notebook_id,
                cell_number,
                output_number,
                output_kind,
            )

            P2AUD_C019_OUTPUT_GROUPS[
                group_key
            ].append(evidence)


for notebook_id, spec in P2AUD_C018_TARGETS.items():
    path = P2AUD_C019_ROOT / spec["path"]

    entry = {
        "notebook": notebook_id,
        "path": spec["path"],
        "read_status": "NOT_READ",
        "sha256_c018": (
            P2AUD_C018_NOTEBOOK_EVIDENCE[
                notebook_id
            ]["current_sha256"]
        ),
        "sha256_current": None,
        "snapshot_status": "NOT_CHECKED",
        "saved_output_groups_with_metrics": 0,
        "execution_reperformed": False,
    }

    try:
        if path.is_symlink():
            entry["read_status"] = "SYMLINK_SKIPPED"

        elif not path.is_file():
            entry["read_status"] = "NOT_FOUND"

        elif not path.resolve().is_relative_to(
            P2AUD_C019_ROOT
        ):
            entry["read_status"] = "OUTSIDE_REPOSITORY"

        elif path.stat().st_size > P2AUD_C019_MAX_BYTES:
            entry["read_status"] = "SIZE_LIMIT"

        else:
            raw = path.read_bytes()
            current_sha = hashlib.sha256(
                raw
            ).hexdigest()

            entry["sha256_current"] = current_sha
            entry["snapshot_status"] = (
                "MATCH_C018"
                if current_sha == entry["sha256_c018"]
                else "CHANGED_SINCE_C018"
            )

            notebook = json.loads(
                raw.decode("utf-8-sig")
            )

            cells = notebook.get("cells", [])

            if not isinstance(cells, list):
                raise ValueError("Invalid notebook cells")

            for cell_number, cell in enumerate(
                cells, start=1
            ):
                if not isinstance(cell, dict):
                    continue

                if cell.get("cell_type") != "code":
                    continue

                source = p2aud_c018_text(
                    cell.get("source", "")
                )

                p2aud_c019_extract_paths(
                    source,
                    f"CODE_CELL_{cell_number}",
                    notebook_id,
                )

                outputs = cell.get("outputs", [])

                if not isinstance(outputs, list):
                    continue

                for output_number, output in enumerate(
                    outputs, start=1
                ):
                    if not isinstance(output, dict):
                        continue

                    output_texts = (
                        p2aud_c018_output_texts(output)
                    )

                    for output_kind, text_value in (
                        output_texts
                    ):
                        if (
                            len(text_value)
                            > P2AUD_C019_MAX_OUTPUT_CHARS
                        ):
                            continue

                        p2aud_c019_extract_output_metrics(
                            text_value,
                            notebook_id,
                            current_sha,
                            cell_number,
                            output_number,
                            output_kind,
                            cell.get("execution_count"),
                        )

                        p2aud_c019_extract_paths(
                            text_value,
                            (
                                f"OUTPUT_CELL_{cell_number}"
                                f"_{output_number}"
                            ),
                            notebook_id,
                        )

            entry["read_status"] = "READ_SUCCESS"

    except (
        OSError, ValueError, TypeError, UnicodeError
    ) as exc:
        entry["read_status"] = "READ_ERROR"
        entry["error_type"] = type(exc).__name__

    entry["saved_output_groups_with_metrics"] = sum(
        key[0] == notebook_id
        for key in P2AUD_C019_OUTPUT_GROUPS
    )

    P2AUD_C019_NOTEBOOK_STATUS[
        notebook_id
    ] = entry


# Deduplicate path references and probe only literal paths.
# Do not infer dynamic paths or scan the whole repository.

P2AUD_C019_UNIQUE_REFERENCES = {}

for record in P2AUD_C019_ARTIFACT_REFERENCES:
    key = (
        record["notebook"],
        record["path_literal"],
    )

    if key not in P2AUD_C019_UNIQUE_REFERENCES:
        P2AUD_C019_UNIQUE_REFERENCES[key] = {
            "notebook": record["notebook"],
            "path_literal": record["path_literal"],
            "first_reference_origin": record["origin"],
            "reference_count": 0,
            "path_status": "NOT_PROBED",
            "file_sha256": None,
        }

    P2AUD_C019_UNIQUE_REFERENCES[
        key
    ]["reference_count"] += 1


P2AUD_C019_ARTIFACT_PROBES = list(
    P2AUD_C019_UNIQUE_REFERENCES.values()
)

for record in P2AUD_C019_ARTIFACT_PROBES:
    relative_path = record["path_literal"]

    if "{" in relative_path or "}" in relative_path:
        record["path_status"] = "DYNAMIC_TEMPLATE"
        continue

    path = P2AUD_C019_ROOT / relative_path

    try:
        if path.is_symlink():
            record["path_status"] = "SYMLINK_SKIPPED"

        elif not path.resolve().is_relative_to(
            P2AUD_C019_ROOT
        ):
            record["path_status"] = "OUTSIDE_REPOSITORY"

        elif path.is_file():
            record["path_status"] = "FILE_FOUND"

            if path.stat().st_size <= 2_000_000:
                record["file_sha256"] = (
                    hashlib.sha256(
                        path.read_bytes()
                    ).hexdigest()
                )
                record["path_status"] = (
                    "FILE_FOUND_HASHED"
                )

        elif path.is_dir():
            record["path_status"] = "DIRECTORY_FOUND"

        else:
            record["path_status"] = (
                "NOT_FOUND_AT_LITERAL_PATH"
            )

    except OSError as exc:
        record["path_status"] = "PROBE_ERROR"
        record["error_type"] = type(exc).__name__


# Rank groups for human review by provenance clues.
# This ranking must never select a winning experiment.

P2AUD_C019_GROUP_SUMMARY = []

for key, matches in P2AUD_C019_OUTPUT_GROUPS.items():
    notebook_id, cell_number, output_number, kind = key

    unique_values = sorted({
        (
            match["metric"],
            match["value"],
        )
        for match in matches
    })

    P2AUD_C019_GROUP_SUMMARY.append({
        "notebook": notebook_id,
        "cell_number": cell_number,
        "output_number": output_number,
        "kind": kind,
        "metric_occurrences": len(matches),
        "distinct_metric_values": len(unique_values),
        "summary_context_occurrences": sum(
            bool(match["summary_word_nearby"])
            for match in matches
        ),
        "split_hint_occurrences": sum(
            bool(match["nearby_split_hints"])
            for match in matches
        ),
        "metric_examples": [
            {"metric": metric, "value": value}
            for metric, value in unique_values[:10]
        ],
        "result_type_verified": False,
    })

P2AUD_C019_GROUP_SUMMARY.sort(
    key=lambda record: (
        -record["summary_context_occurrences"],
        -record["split_hint_occurrences"],
        record["notebook"],
        record["cell_number"],
        record["output_number"],
    )
)

P2AUD_C019_MATRIX = copy.deepcopy(
    P2AUD_C018_MATRIX
)

P2AUD_C019_NB22 = next(
    row
    for row in P2AUD_C019_MATRIX
    if row["notebook_id"] == "NB22"
)

P2AUD_C019_NB22["metric_provenance_review"] = {
    "source_notebooks": P2AUD_C019_NOTEBOOK_STATUS,
    "metric_context_count": len(
        P2AUD_C019_METRIC_CONTEXT
    ),
    "metric_output_group_count": len(
        P2AUD_C019_OUTPUT_GROUPS
    ),
    "literal_artifact_reference_count": len(
        P2AUD_C019_ARTIFACT_PROBES
    ),
    "metric_population_adjudicated": False,
    "latest_selected_configuration_identified": False,
    "formal_acceptance": "NOT_VERIFIED",
}

P2AUD_C019_REPORT = {
    "schema_version": (
        "P2AUD_NB22_METRIC_CONTEXT_AND_ARTIFACT_PROVENANCE_V1"
    ),
    "cell_id": "P2AUD-C019",
    "action": 19,
    "status": (
        "P2AUD_C019_PROVENANCE_CANDIDATES_INDEXED"
        "_ADJUDICATION_PENDING"
    ),
    "notebook_status": (
        P2AUD_C019_NOTEBOOK_STATUS
    ),
    "metric_occurrences_with_line_context": len(
        P2AUD_C019_METRIC_CONTEXT
    ),
    "metric_output_groups": len(
        P2AUD_C019_OUTPUT_GROUPS
    ),
    "output_group_review_candidates": (
        P2AUD_C019_GROUP_SUMMARY[:18]
    ),
    "artifact_literal_references": len(
        P2AUD_C019_ARTIFACT_PROBES
    ),
    "artifact_probe_status_counts": dict(
        Counter(
            record["path_status"]
            for record in P2AUD_C019_ARTIFACT_PROBES
        )
    ),
    "artifact_probe_examples": (
        P2AUD_C019_ARTIFACT_PROBES[:20]
    ),
    "metric_context_examples": {
        notebook_id: [
            {
                "cell_number": item["cell_number"],
                "output_number": item["output_number"],
                "output_line": item["output_line"],
                "metric": item["metric"],
                "value": item["value"],
                "split_hints": (
                    item["nearby_split_hints"]
                ),
                "summary_hint": (
                    item["summary_word_nearby"]
                ),
                "context_excerpt": (
                    item["context_excerpt"]
                ),
            }
            for item in P2AUD_C019_METRIC_CONTEXT
            if item["notebook"] == notebook_id
        ][:8]
        for notebook_id in P2AUD_C018_TARGETS
    },
    "matrix_rows_preserved": len(
        P2AUD_C019_MATRIX
    ),
    "metrics_formally_adjudicated": 0,
    "release_accepted": False,
    "files_written": 0,
    "new_api_calls": 0,
    "git_operations": 0,
    "next_step": (
        "VERIFY_CANDIDATE_AGGREGATE_REPORTS"
        "_AND_RETURN_TO_FULL_AS_BUILT_MATRIX"
    ),
}

assert len(P2AUD_C019_MATRIX) == 24

print(
    json.dumps(
        P2AUD_C019_REPORT,
        ensure_ascii=False,
        indent=2,
    )
)
````

## P2AUD-C020 — تثبیت دفتر شواهد 22e و 22f

### هدف

جمع‌بندی شواهد موجود و اتصال گزارش‌های نسخه‌دار به NB22، بدون ادامهٔ جست‌وجوی گسترده.

### کارها

````python
# P2AUD-C020 — Consolidate NB22 evidence and return to As-Built
# READ_ONLY | NO_API | NO_FILE_WRITES | NO_GIT_WRITES

import copy
import json
import hashlib
import re
from pathlib import Path
from collections import Counter, defaultdict

assert P2AUD_C019_REPORT["status"] == (
    "P2AUD_C019_PROVENANCE_CANDIDATES_INDEXED"
    "_ADJUDICATION_PENDING"
)
assert len(P2AUD_C019_MATRIX) == 24

P2AUD_C020_ROOT = Path(P2AUD_REPO_ROOT).resolve()

# 1. Separate metric occurrences from distinct score pairs.

P2AUD_C020_F_HITS = sorted(
    [
        item for item in P2AUD_C019_METRIC_CONTEXT
        if item["notebook"] == "NB22F_V1"
    ],
    key=lambda item: (
        item["cell_number"],
        item["output_number"],
        item["output_line"],
    ),
)

P2AUD_C020_F_LINE_INDEX = defaultdict(list)

for item in P2AUD_C020_F_HITS:
    key = (
        item["cell_number"],
        item["output_number"],
        item["output_kind"],
        item["output_line"],
    )
    P2AUD_C020_F_LINE_INDEX[key].append(item)

P2AUD_C020_F_PAIRS = []

for recall in P2AUD_C020_F_HITS:
    if recall["metric"] != "RECALL_AT_10":
        continue

    next_key = (
        recall["cell_number"],
        recall["output_number"],
        recall["output_kind"],
        recall["output_line"] + 1,
    )

    ndcg_matches = [
        item for item in P2AUD_C020_F_LINE_INDEX[next_key]
        if item["metric"] == "NDCG_AT_10"
    ]

    if len(ndcg_matches) != 1:
        continue

    ndcg = ndcg_matches[0]

    P2AUD_C020_F_PAIRS.append({
        "cell_number": recall["cell_number"],
        "output_number": recall["output_number"],
        "recall_line": recall["output_line"],
        "ndcg_line": ndcg["output_line"],
        "recall_at_10": recall["value"],
        "ndcg_at_10": ndcg["value"],
        "population": "NOT_VERIFIED",
        "configuration": "NOT_VERIFIED",
        "acceptance": "NOT_EVALUATED",
    })

P2AUD_C020_F_DISTINCT_PAIRS = {}

for pair in P2AUD_C020_F_PAIRS:
    key = (
        pair["recall_at_10"],
        pair["ndcg_at_10"],
    )

    if key not in P2AUD_C020_F_DISTINCT_PAIRS:
        P2AUD_C020_F_DISTINCT_PAIRS[key] = {
            "recall_at_10": pair["recall_at_10"],
            "ndcg_at_10": pair["ndcg_at_10"],
            "first_cell": pair["cell_number"],
            "first_output_line": pair["recall_line"],
            "occurrences": 0,
            "evaluation_role": "NOT_VERIFIED",
        }

    P2AUD_C020_F_DISTINCT_PAIRS[
        key
    ]["occurrences"] += 1


# 2. Out-of-range matches need contextual review.
# They are not automatically invalid experiments.

P2AUD_C020_OUT_OF_RANGE = [
    item
    for item in P2AUD_C019_METRIC_CONTEXT
    if not 0.0 <= item["value"] <= 1.0
]


# 3. Choose a bounded set of report artifacts, using
# paths actually discovered by C019.

P2AUD_C020_CANDIDATES = [
    item
    for item in P2AUD_C019_ARTIFACT_PROBES
    if (
        item["path_status"] == "FILE_FOUND_HASHED"
        and item["path_literal"].endswith(".json")
        and item["path_literal"].startswith(
            "reports/phase2/quality_improvement/NB22"
        )
    )
]

P2AUD_C020_REPORT_NAME_RE = re.compile(
    r"final|summary|consolidat|benchmark|"
    r"selection|selected|evaluation|"
    r"manifest|readiness|acceptance|gate",
    re.IGNORECASE,
)


def p2aud_c020_priority(item):
    path = item["path_literal"]
    filename = Path(path).name

    return (
        0 if P2AUD_C020_REPORT_NAME_RE.search(filename) else 1,
        len(path),
        path,
    )


P2AUD_C020_SELECTED = []
P2AUD_C020_SELECTED_PATHS = set()

for directory_marker in (
    "/NB22F",
    "/NB22E.lab.v2/",
):
    matching = sorted(
        [
            item for item in P2AUD_C020_CANDIDATES
            if directory_marker in item["path_literal"]
        ],
        key=p2aud_c020_priority,
    )

    for item in matching[:6]:
        path = item["path_literal"]

        if path not in P2AUD_C020_SELECTED_PATHS:
            P2AUD_C020_SELECTED.append(item)
            P2AUD_C020_SELECTED_PATHS.add(path)


# 4. Inspect only bounded metadata and metric fields.
# No candidate can approve a release automatically.

P2AUD_C020_METRIC_KEY_RE = re.compile(
    r"^(?:recall_at_10|ndcg_at_10)$",
    re.IGNORECASE,
)

P2AUD_C020_META_KEY_RE = re.compile(
    r"status|split|population|dataset|"
    r"version|source|candidate|selected|"
    r"config|release|gate",
    re.IGNORECASE,
)


def p2aud_c020_fields(data, max_nodes=3000):
    metric_fields = []
    metadata_fields = []
    nodes = 0
    truncated = False

    def walk(value, prefix="", depth=0):
        nonlocal nodes, truncated

        if nodes >= max_nodes or depth > 6:
            truncated = True
            return

        nodes += 1

        if isinstance(value, dict):
            for key, child in value.items():
                path = (
                    f"{prefix}.{key}"
                    if prefix else str(key)
                )

                if (
                    P2AUD_C020_METRIC_KEY_RE.fullmatch(
                        str(key)
                    )
                    and isinstance(child, (int, float))
                    and not isinstance(child, bool)
                ):
                    if len(metric_fields) < 12:
                        metric_fields.append({
                            "field": path,
                            "value": child,
                        })

                if (
                    P2AUD_C020_META_KEY_RE.search(str(key))
                    and isinstance(
                        child, (str, int, float, bool)
                    )
                    and len(metadata_fields) < 12
                ):
                    metadata_fields.append({
                        "field": path,
                        "value": str(child)[:100],
                    })

                if isinstance(child, (dict, list)):
                    walk(child, path, depth + 1)

                if nodes >= max_nodes:
                    truncated = True
                    break

        elif isinstance(value, list):
            for index, child in enumerate(value[:25]):
                if isinstance(child, (dict, list)):
                    walk(
                        child,
                        f"{prefix}[{index}]",
                        depth + 1,
                    )

                if nodes >= max_nodes:
                    truncated = True
                    break

            if len(value) > 25:
                truncated = True

    walk(data)

    return {
        "metric_fields": metric_fields,
        "metadata_fields": metadata_fields,
        "nodes_inspected": nodes,
        "inspection_truncated": truncated,
    }


P2AUD_C020_REPORT_ARTIFACTS = []

for candidate in P2AUD_C020_SELECTED:
    relative_path = candidate["path_literal"]
    path = P2AUD_C020_ROOT / relative_path

    record = {
        "path": relative_path,
        "status": "NOT_READ",
        "sha256_c019": candidate["file_sha256"],
        "sha256_current": None,
        "snapshot_status": "NOT_CHECKED",
        "metric_fields": [],
        "metadata_fields": [],
        "inspection_truncated": False,
        "acceptance_verified": False,
    }

    try:
        if path.is_symlink():
            record["status"] = "SYMLINK_SKIPPED"

        elif not path.is_file():
            record["status"] = "FILE_NOT_FOUND"

        elif not path.resolve().is_relative_to(
            P2AUD_C020_ROOT
        ):
            record["status"] = "OUTSIDE_REPOSITORY"

        elif path.stat().st_size > 1_000_000:
            record["status"] = "SIZE_LIMIT"

        else:
            raw = path.read_bytes()
            current_sha = hashlib.sha256(
                raw
            ).hexdigest()

            record["sha256_current"] = current_sha
            record["snapshot_status"] = (
                "MATCH_C019"
                if current_sha == record["sha256_c019"]
                else "CHANGED_SINCE_C019"
            )

            data = json.loads(
                raw.decode("utf-8-sig")
            )

            fields = p2aud_c020_fields(data)

            record.update({
                "status": "READ_SUCCESS",
                **fields,
            })

    except (OSError, ValueError, UnicodeError) as exc:
        record["status"] = "READ_ERROR"
        record["error_type"] = type(exc).__name__

    P2AUD_C020_REPORT_ARTIFACTS.append(record)


# 5. Attach the bounded evidence to the 24-row matrix.

P2AUD_C020_MATRIX = copy.deepcopy(
    P2AUD_C019_MATRIX
)

P2AUD_C020_NB22 = next(
    row for row in P2AUD_C020_MATRIX
    if row["notebook_id"] == "NB22"
)

P2AUD_C020_NB22["bounded_metric_review"] = {
    "nb22f_distinct_score_pairs": list(
        P2AUD_C020_F_DISTINCT_PAIRS.values()
    ),
    "out_of_range_text_matches": len(
        P2AUD_C020_OUT_OF_RANGE
    ),
    "versioned_reports_inspected": (
        P2AUD_C020_REPORT_ARTIFACTS
    ),
    "historical_metrics_preserved": True,
    "latest_final_configuration_verified": False,
    "independent_holdout_verified": False,
    "quality_stability_verified": False,
    "release_accepted": False,
    "review_scope": (
        "BOUNDED_EVIDENCE_INDEXING_COMPLETE"
        "_FORMAL_ACCEPTANCE_OPEN"
    ),
}

P2AUD_C020_REPORT = {
    "schema_version": (
        "P2AUD_NB22_BOUNDED_EVIDENCE_SUMMARY_V1"
    ),
    "cell_id": "P2AUD-C020",
    "action": 20,
    "status": (
        "P2AUD_C020_NB22_EVIDENCE_SUMMARIZED"
        "_RETURN_TO_AS_BUILT"
    ),
    "notebook_snapshots_verified": 2,
    "nb22f_score_pair_occurrences": len(
        P2AUD_C020_F_PAIRS
    ),
    "nb22f_distinct_score_pairs": list(
        P2AUD_C020_F_DISTINCT_PAIRS.values()
    ),
    "out_of_range_metric_text_matches": len(
        P2AUD_C020_OUT_OF_RANGE
    ),
    "out_of_range_examples": [
        {
            "notebook": item["notebook"],
            "cell": item["cell_number"],
            "metric": item["metric"],
            "value": item["value"],
        }
        for item in P2AUD_C020_OUT_OF_RANGE[:8]
    ],
    "matching_versioned_report_candidates": len(
        P2AUD_C020_CANDIDATES
    ),
    "targeted_reports_read": len([
        item for item in P2AUD_C020_REPORT_ARTIFACTS
        if item["status"] == "READ_SUCCESS"
    ]),
    "targeted_report_summary": [
        {
            "path": item["path"],
            "status": item["status"],
            "snapshot_status": item["snapshot_status"],
            "metric_fields": item["metric_fields"][:6],
            "metadata_fields": item["metadata_fields"][:6],
            "inspection_truncated": (
                item["inspection_truncated"]
            ),
        }
        for item in P2AUD_C020_REPORT_ARTIFACTS
    ],
    "matrix_rows_preserved": len(
        P2AUD_C020_MATRIX
    ),
    "quality_gate_newly_accepted": False,
    "release_accepted": False,
    "files_written": 0,
    "new_api_calls": 0,
    "git_operations": 0,
    "next_step": (
        "FINALIZE_ALL_NOTEBOOK_AS_BUILT"
        "_ROADMAP_DIFFERENCE_MATRIX"
    ),
}

assert len(P2AUD_C020_MATRIX) == 24

print(
    json.dumps(
        P2AUD_C020_REPORT,
        ensure_ascii=False,
        indent=2,
    )
)
````

````python
# P2AUD-C022 — Documentary reconciliation of five roadmap differences
# READ_ONLY | NO_API | NO_FILE_WRITES | NO_GIT_WRITES

import copy
import json
import re
import hashlib
from pathlib import Path
from collections import Counter

assert P2AUD_C021_REPORT["status"] == (
    "P2AUD_C021_AS_BUILT_DRAFT_CREATED"
    "_FINAL_RECONCILIATION_OPEN"
)
assert len(P2AUD_C021_AS_BUILT_MAP) == 24
assert len(P2AUD_C021_PACKAGE_CROSSWALK) == 14
assert len(P2AUD_C021_REVIEW_CASES) == 5

P2AUD_C022_ROOT = Path(P2AUD_REPO_ROOT).resolve()

P2AUD_C022_SOURCE_PATHS = {
    "MODULAR": (
        "data/Road_Maps/"
        "RAG_finance_phase2_modular_notebooks_execution_plan.md"
    ),
    "OPERATIONAL": (
        "data/Road_Maps/"
        "RAG_finance_phase2_operational_roadmap.md"
    ),
}

P2AUD_C022_SOURCES = {}

for source_id, relative_path in P2AUD_C022_SOURCE_PATHS.items():
    path = P2AUD_C022_ROOT / relative_path

    record = {
        "relative_path": relative_path,
        "read_status": "NOT_READ",
        "sha256": None,
        "line_count": 0,
        "lines": [],
    }

    try:
        if path.is_symlink():
            record["read_status"] = "SYMLINK_SKIPPED"
        elif not path.is_file():
            record["read_status"] = "NOT_FOUND"
        elif path.stat().st_size > 2_000_000:
            record["read_status"] = "SIZE_LIMIT"
        else:
            raw = path.read_bytes()
            lines = raw.decode("utf-8-sig").splitlines()

            record.update({
                "read_status": "READ_SUCCESS",
                "sha256": hashlib.sha256(raw).hexdigest(),
                "line_count": len(lines),
                "lines": lines,
            })
    except (OSError, UnicodeError) as exc:
        record["read_status"] = "READ_ERROR"
        record["error_type"] = type(exc).__name__

    P2AUD_C022_SOURCES[source_id] = record


# Each anchor refers to a concrete statement in the source,
# not an invented requirement or a keyword count.

P2AUD_C022_ANCHORS = {
    "P2AUD-RV-001": [
        ("MODULAR", "CLAIM_COVERAGE",
         r"Claim citation coverage"),
        ("MODULAR", "CITATION_CORRECTNESS",
         r"Citation correctness"),
        ("OPERATIONAL", "CITATION_COVERAGE",
         r"Citation coverage"),
    ],
    "P2AUD-RV-002": [
        ("MODULAR", "UNIFIED_SCHEMA_NB07",
         r"## Notebook 07"),
        ("OPERATIONAL", "EARLY_CONTRACT_P2_02",
         r"بسته P2-02.*Data Contract"),
        ("OPERATIONAL", "CONTRACT_PRECEDES_INGESTION",
         r"P2-03 تا P2-05.*Contract"),
    ],
    "P2AUD-RV-003": [
        ("MODULAR", "NB11_BEFORE_NB12",
         r"L --> M\[12 Golden Dataset\]"),
        ("OPERATIONAL", "GOLDEN_CHUNKING_PARALLEL",
         r"P2-08 Golden Dataset.*P2-07 Chunking"),
    ],
    "P2AUD-RV-004": [
        ("MODULAR", "NB10_TAXONOMY",
         r"## Notebook 10"),
        ("MODULAR", "TAXONOMY_EVIDENCE_OWNER",
         r"مهندسی دانش.*Taxonomy.*10"),
        ("OPERATIONAL", "P2_14_DOCUMENTATION",
         r"بسته P2-14.*مستندات دانش فنی"),
    ],
    "P2AUD-RV-005": [
        ("MODULAR", "NB23_HANDOFF",
         r"## Notebook 23"),
        ("MODULAR", "HANDOFF_CONTRACT_OUTPUT",
         r"phase3_handoff_contract\.json"),
        ("OPERATIONAL", "GLOBAL_PHASE3_GATE",
         r"Gate نهایی و تحویل به فاز ۳"),
        ("OPERATIONAL", "INPUT_OUTPUT_CONTRACT",
         r"RAG input/output contract"),
    ],
}

P2AUD_C022_EXPLANATIONS = {
    "P2AUD-RV-001": (
        "اختلاف عددی مستند میان Coverage ماژولار "
        "و معیار اولیهٔ عملیاتی؛ تعیین دامنه و "
        "معیار نهایی به قرارداد فریز‌شده وابسته است."
    ),
    "P2AUD-RV-002": (
        "P2-02 قراردادهای پایه را پیش از Ingestion "
        "قرار می‌دهد؛ NB07 یکپارچه‌سازی Schema "
        "منابع را توضیح می‌دهد. ترتیب اجرای واقعی "
        "به‌تنهایی از این دو نقشه اثبات نمی‌شود."
    ),
    "P2AUD-RV-003": (
        "نقشهٔ ماژولار ترتیب مفهومی NB11 و NB12 "
        "را نشان می‌دهد؛ نقشهٔ عملیاتی امکان "
        "پیشبرد موازی P2-07 و P2-08 را مطرح می‌کند."
    ),
    "P2AUD-RV-004": (
        "NB10 در نقشهٔ ماژولار مسئول مستقیم "
        "Taxonomy است؛ در نگاشت ۱۴ بسته، مالک "
        "عملیاتی اصلی مستقل مشخص نشده است. "
        "این موضوع به‌تنهایی نبود پیاده‌سازی نیست."
    ),
    "P2AUD-RV-005": (
        "NB23 مسئول قرارداد Handoff در نقشهٔ "
        "ماژولار است؛ نقشهٔ عملیاتی تعهدات "
        "تحویل فاز سوم را در Gate سراسری دارد. "
        "Draft ساخته‌شده از پذیرش رسمی جداست."
    ),
}

P2AUD_C022_DIGITS = str.maketrans(
    "۰۱۲۳۴۵۶۷۸۹٠١٢٣٤٥٦٧٨٩٪٫",
    "01234567890123456789%."
)


def p2aud_c022_find(source_id, label, pattern):
    source = P2AUD_C022_SOURCES[source_id]

    if source["read_status"] != "READ_SUCCESS":
        return {
            "source": source_id,
            "anchor": label,
            "found": False,
            "source_read_status": source["read_status"],
        }

    regex = re.compile(pattern, re.IGNORECASE)

    for index, line in enumerate(source["lines"], start=1):
        if regex.search(line):
            return {
                "source": source_id,
                "anchor": label,
                "found": True,
                "line": index,
                "excerpt": line.strip()[:260],
                "source_sha256": source["sha256"],
            }

    return {
        "source": source_id,
        "anchor": label,
        "found": False,
        "source_read_status": "READ_SUCCESS",
    }


P2AUD_C022_REVIEW_CASES = []

for previous in P2AUD_C021_REVIEW_CASES:
    review_id = previous["review_id"]

    anchors = [
        p2aud_c022_find(source_id, label, pattern)
        for source_id, label, pattern
        in P2AUD_C022_ANCHORS[review_id]
    ]

    all_anchors_found = all(
        item["found"] for item in anchors
    )

    if not all_anchors_found:
        documentary_status = (
            "SOURCE_EVIDENCE_INCOMPLETE"
        )

    elif review_id == "P2AUD-RV-001":
        by_label = {
            item["anchor"]: item["excerpt"].translate(
                P2AUD_C022_DIGITS
            )
            for item in anchors
        }

        modular_90 = bool(re.search(
            r"0\.90\b|90\s*%",
            by_label["CLAIM_COVERAGE"]
        ))
        operational_95 = bool(re.search(
            r"0\.95\b|95\s*%",
            by_label["CITATION_COVERAGE"]
        ))

        documentary_status = (
            "NUMERIC_THRESHOLD_DIFFERENCE_DOCUMENTED"
            if modular_90 and operational_95
            else "THRESHOLD_DETAILS_REQUIRE_REVIEW"
        )

    elif review_id == "P2AUD-RV-002":
        documentary_status = (
            "BASE_CONTRACT_VS_UNIFIED_SCHEMA_SCOPE"
        )

    elif review_id == "P2AUD-RV-003":
        documentary_status = (
            "CONCEPTUAL_SEQUENCE_VS_ALLOWED_PARALLELISM"
        )

    elif review_id == "P2AUD-RV-004":
        documentary_status = (
            "MODULAR_OWNER_VS_OPERATIONAL_MAPPING_GAP"
        )

    else:
        documentary_status = (
            "NOTEBOOK_HANDOFF_VS_GLOBAL_DELIVERY_GATE"
        )

    record = {
        "review_id": review_id,
        "topic": previous["topic"],
        "source_anchors": anchors,
        "all_source_anchors_found": all_anchors_found,
        "documentary_status": documentary_status,
        "interpretation": (
            P2AUD_C022_EXPLANATIONS[review_id]
            if all_anchors_found
            else "شواهد کافی برای تفسیر مستند پیدا نشد."
        ),
        "affected_notebooks": previous["notebooks"],
        "related_packages": previous[
            "related_operational_packages"
        ],
        "implementation_failure_proven": False,
        "formal_requirement_decision_approved": False,
        "needs_final_contract_or_owner_review": (
            review_id == "P2AUD-RV-001"
            or not all_anchors_found
        ),
    }

    P2AUD_C022_REVIEW_CASES.append(record)


# Update, without replacing, the existing 24-row As-Built.

P2AUD_C022_AS_BUILT_MAP = copy.deepcopy(
    P2AUD_C021_AS_BUILT_MAP
)

P2AUD_C022_CASE_INDEX = {
    case["review_id"]: case
    for case in P2AUD_C022_REVIEW_CASES
}

for entry in P2AUD_C022_AS_BUILT_MAP:
    entry["roadmap_documentary_reconciliation"] = [
        {
            "review_id": review_id,
            "documentary_status": (
                P2AUD_C022_CASE_INDEX[
                    review_id
                ]["documentary_status"]
            ),
            "formal_decision_approved": False,
        }
        for review_id in entry["review_case_ids"]
    ]

    entry["implementation_status_preserved"] = True
    entry["acceptance_status_preserved"] = True

assert len(P2AUD_C022_AS_BUILT_MAP) == 24


# Produce a revised complete Markdown map in memory.

P2AUD_C022_MAP_PREFIX = (
    P2AUD_C021_MAP_MARKDOWN.split(
        "## اختلاف‌ها و ابهام‌های دو نقشه"
    )[0]
)

P2AUD_C022_LINES = [
    P2AUD_C022_MAP_PREFIX.rstrip(),
    "",
    "## تطبیق مستند پنج اختلاف نقشه",
    "",
    "| مورد | نتیجهٔ بررسی متن | تصمیم رسمی |",
    "|---|---|---|",
]

for case in P2AUD_C022_REVIEW_CASES:
    P2AUD_C022_LINES.append(
        "| "
        + case["review_id"]
        + " | "
        + case["documentary_status"]
        + " | باز / تأیید نشده |"
    )

P2AUD_C022_LINES.extend([
    "",
    "### توضیح اختلاف‌ها",
    "",
])

for case in P2AUD_C022_REVIEW_CASES:
    P2AUD_C022_LINES.extend([
        f'**{case["review_id"]}:** '
        + case["interpretation"],
        "",
    ])

P2AUD_C022_LINES.extend([
    "### محدودیت‌های As-Built",
    "",
    "- وضعیت تحویل بر گزارش مالک پروژه و "
    "موجودی شواهد متکی است؛ پذیرش مستقل "
    "همهٔ الزامات هنوز اثبات نشده است.",
    "- اختلاف قرارداد Citation باید با "
    "نسخهٔ فریز‌شده تعیین تکلیف شود.",
    "- نتایج بهتر NB22e در Validation "
    "به معنی پذیرش پایدار Release نیست.",
    "- تحویل Draft نوت‌بوک NB23 از "
    "Handoff رسمی به فاز سوم جداست.",
    "- صفر گزارش در فهرست نام‌محور، "
    "به معنی انجام‌نشدن یک نوت‌بوک نیست.",
])

P2AUD_C022_MAP_MARKDOWN = "\n".join(
    P2AUD_C022_LINES
)

try:
    from IPython.display import Markdown, display
    display(Markdown(P2AUD_C022_MAP_MARKDOWN))
except ImportError:
    print(P2AUD_C022_MAP_MARKDOWN)


P2AUD_C022_STATUS_COUNTS = Counter(
    case["documentary_status"]
    for case in P2AUD_C022_REVIEW_CASES
)

P2AUD_C022_REPORT = {
    "schema_version": (
        "P2AUD_PHASE2_DOCUMENTARY_ROADMAP_RECONCILIATION_V1"
    ),
    "cell_id": "P2AUD-C022",
    "action": 22,
    "status": (
        "P2AUD_C022_DOCUMENTARY_RECONCILIATION_COMPLETE"
        "_FORMAL_DECISIONS_OPEN"
    ),
    "roadmap_sources": {
        key: {
            "path": value["relative_path"],
            "read_status": value["read_status"],
            "sha256": value["sha256"],
            "line_count": value["line_count"],
        }
        for key, value in P2AUD_C022_SOURCES.items()
    },
    "notebooks_preserved": len(
        P2AUD_C022_AS_BUILT_MAP
    ),
    "operational_packages_preserved": len(
        P2AUD_C021_PACKAGE_CROSSWALK
    ),
    "review_cases_examined": len(
        P2AUD_C022_REVIEW_CASES
    ),
    "review_case_status_counts": dict(
        sorted(P2AUD_C022_STATUS_COUNTS.items())
    ),
    "review_cases": [
        {
            "id": case["review_id"],
            "status": case["documentary_status"],
            "source_anchors": case["source_anchors"],
            "interpretation": case["interpretation"],
            "formal_decision": "NOT_APPROVED",
        }
        for case in P2AUD_C022_REVIEW_CASES
    ],
    "formal_roadmap_decisions_approved": 0,
    "new_quality_gates_approved": 0,
    "new_release_acceptance": False,
    "files_written": 0,
    "api_calls": 0,
    "git_operations": 0,
    "next_step": (
        "PREPARE_VERSIONED_AS_BUILT_DELIVERABLE"
        "_WITH_EVIDENCE_AND_ACCEPTANCE_LIMITS"
    ),
}

print("\n========== P2AUD-C022 SUMMARY ==========\n")
print(json.dumps(
    P2AUD_C022_REPORT,
    ensure_ascii=False,
    indent=2,
))
````

````python
# P2AUD-C023 — Consolidated Phase 2 As-Built deliverable
# READ_ONLY | NO_API | NO_FILE_WRITES | NO_GIT_WRITES

import json
import hashlib
from collections import Counter

assert P2AUD_C022_REPORT["status"] == (
    "P2AUD_C022_DOCUMENTARY_RECONCILIATION_COMPLETE"
    "_FORMAL_DECISIONS_OPEN"
)
assert len(P2AUD_C022_AS_BUILT_MAP) == 24
assert len(P2AUD_C021_PACKAGE_CROSSWALK) == 14
assert len(P2AUD_C022_REVIEW_CASES) == 5

def p2aud_c023_md(value):
    return str(value).replace("|", "\\|").replace("\n", " ")

def p2aud_c023_packages(values):
    return ", ".join(values) if values else "بدون مالک اصلی مستقل"

P2AUD_C023_STATUS_FA = {
    "IMPLEMENTED_ACCORDING_TO_PROJECT_OWNER":
        "پیاده‌سازی‌شده طبق گزارش مالک؛ پذیرش مستقل بررسی‌نشده",
    "CHECKPOINT_IMPLEMENTED_QUALITY_ACCEPTANCE_OPEN":
        "Checkpoint انجام‌شده؛ پذیرش کیفیت باز",
    "DRAFT_CONTRACT_IMPLEMENTED":
        "قرارداد Draft انجام‌شده؛ تحویل رسمی باز",
}

P2AUD_C023_LINES = [
    "# نقشهٔ واقعی اجرای فاز دوم RAG_finance",
    "",
    "**نسخه:** P2AUD-AS-BUILT-DRAFT-V1",
    "",
    "**وضعیت:** قابل بازبینی؛ فاقد پذیرش رسمی Release",
    "",
    "## ۱. خلاصهٔ اجرایی",
    "",
    "فاز دوم طبق گزارش مالک پروژه از نظر پیاده‌سازی اصلی "
    "عملاً انجام شده است. ممیزی، وجود ۲۴ نوت‌بوک اصلی "
    "و ۲۴ نوت‌بوک تکمیلی مرتبط را ثبت کرده است.",
    "",
    "این سند سه موضوع را جدا نگه می‌دارد: "
    "**پیاده‌سازی، شواهد اجرا و پذیرش نهایی**.",
    "",
    "بهبود کیفیت بازیابی در NB22 همچنان ادامه دارد. "
    "قرارداد Draft در NB23 ساخته شده، اما تحویل رسمی "
    "فاز سوم وابسته به پذیرش Release است.",
    "",
    "## ۲. منابع و نسخهٔ نقشه‌ها",
    "",
]

for source_id in ("MODULAR", "OPERATIONAL"):
    record = P2AUD_C022_SOURCES[source_id]
    P2AUD_C023_LINES.extend([
        f"- **{source_id}:** `{record['relative_path']}`",
        f"  - SHA-256: `{record['sha256']}`",
        f"  - وضعیت خواندن: `{record['read_status']}`",
    ])

P2AUD_C023_LINES.extend([
    "",
    "## ۳. نقشهٔ اجرایی NB00 تا NB23",
    "",
])

P2AUD_C023_INDEX = {
    row["notebook_id"]: row
    for row in P2AUD_C022_AS_BUILT_MAP
}

for stage in P2AUD_C021_STAGE_DEFINITIONS:
    P2AUD_C023_LINES.extend([
        f"### {stage['id']} — {stage['title']}",
        "",
        "| NB | مسئولیت | بستهٔ اصلی | فرعی | "
        "نامزد گزارش | وضعیت |",
        "|---|---|---|---:|---:|---|",
    ])

    for notebook_id in stage["notebooks"]:
        item = P2AUD_C023_INDEX[notebook_id]

        values = [
            notebook_id,
            item["planned_responsibility"],
            p2aud_c023_packages(
                item["operational_primary_packages"]
            ),
            item["supplementary_notebook_count"],
            item["identified_named_report_candidates"],
            P2AUD_C023_STATUS_FA.get(
                item["delivery_status"],
                item["delivery_status"],
            ),
        ]

        P2AUD_C023_LINES.append(
            "| " + " | ".join(
                p2aud_c023_md(value)
                for value in values
            ) + " |"
        )

    P2AUD_C023_LINES.append("")


P2AUD_C023_LINES.extend([
    "## ۴. نگاشت ۱۴ بستهٔ عملیاتی",
    "",
    "| بسته | نوت‌بوک‌های اصلی | نوت‌بوک‌های پشتیبان |",
    "|---|---|---|",
])

for package in P2AUD_C021_PACKAGE_CROSSWALK:
    P2AUD_C023_LINES.append(
        "| "
        + " | ".join([
            package["package_id"],
            p2aud_c023_packages(
                package["primary_notebooks"]
            ),
            p2aud_c023_packages(
                package["support_notebooks"]
            ),
        ])
        + " |"
    )


P2AUD_C023_LINES.extend([
    "",
    "## ۵. اختلاف‌های دو نقشه",
    "",
    "| شناسه | موضوع | نتیجهٔ مستند | تصمیم رسمی |",
    "|---|---|---|---|",
])

for case in P2AUD_C022_REVIEW_CASES:
    P2AUD_C023_LINES.append(
        "| "
        + " | ".join([
            case["review_id"],
            p2aud_c023_md(case["topic"]),
            case["documentary_status"],
            "باز",
        ])
        + " |"
    )

P2AUD_C023_LINES.extend([
    "",
    "### منشأ و توضیح اختلاف‌ها",
    "",
])

for case in P2AUD_C022_REVIEW_CASES:
    P2AUD_C023_LINES.extend([
        f"**{case['review_id']} — {case['topic']}**",
        "",
        case["interpretation"],
        "",
    ])

    for anchor in case["source_anchors"]:
        if anchor.get("found"):
            P2AUD_C023_LINES.append(
                f"- `{anchor['source']}` خط "
                f"{anchor['line']}: "
                f"{p2aud_c023_md(anchor['excerpt'])}"
            )
        else:
            P2AUD_C023_LINES.append(
                f"- `{anchor['source']}` / "
                f"`{anchor['anchor']}`: شاهد مستقیم پیدا نشد."
            )

    P2AUD_C023_LINES.append("")


# NB22 historical quality and newer experimental evidence.

P2AUD_C023_QUALITY = P2AUD_C017_HISTORICAL_QUALITY
P2AUD_C023_GATE = (
    P2AUD_C023_QUALITY["recorded_gate_counts"]
)

P2AUD_C023_LINES.extend([
    "## ۶. وضعیت کیفیت و پذیرش",
    "",
    "### NB22 — بازیابی و Release",
    "",
    "| سنجه | نتیجهٔ تاریخی Test | هدف |",
    "|---|---:|---:|",
    "| Recall@10 | "
    f"{P2AUD_C023_QUALITY['historical_test_recall_at_10']:.4f}"
    " | "
    f"{P2AUD_C023_QUALITY['frozen_recall_at_10_min']:.2f}"
    " |",
    "| nDCG@10 | "
    f"{P2AUD_C023_QUALITY['historical_test_ndcg_at_10']:.4f}"
    " | "
    f"{P2AUD_C023_QUALITY['frozen_ndcg_at_10_min']:.2f}"
    " |",
    "",
    "**Gate تاریخی:** "
    f"{P2AUD_C023_GATE.get('PASS', 0)} PASS، "
    f"{P2AUD_C023_GATE.get('FAIL', 0)} FAIL، "
    f"{P2AUD_C023_GATE.get('BLOCKED', 0)} BLOCKED.",
    "",
    "آزمایش‌های جدید 22e روی Validation به نتایج "
    "بهتری رسیده‌اند؛ اما ثبت نتیجهٔ پایدار روی "
    "نسخهٔ پذیرفته‌شده هنوز باز است.",
    "",
    "### NB23 — تحویل فاز سوم",
    "",
    "قرارداد Draft و آزمون‌های ساختاری تهیه شده‌اند. "
    "پذیرش رسمی Handoff وابسته به Release معتبر "
    "و اتصال نسخه‌های واقعی Producer است.",
    "",
    "## ۷. دفتر نسخه‌های تکمیلی",
    "",
])

for item in P2AUD_C022_AS_BUILT_MAP:
    variants = item["supplementary_notebook_paths"]

    if not variants:
        continue

    P2AUD_C023_LINES.extend([
        f"### {item['notebook_id']}",
        "",
    ])

    for path in variants:
        P2AUD_C023_LINES.append(f"- `{path}`")

    P2AUD_C023_LINES.append("")


P2AUD_C023_LINES.extend([
    "### فایل‌های کمکی خارج از دسته‌بندی NB00–NB23",
    "",
])

for path in P2AUD_C015_UNCLASSIFIED_VARIANTS:
    P2AUD_C023_LINES.append(f"- `{path}`")


P2AUD_C023_LINES.extend([
    "",
    "## ۸. کارهای باقی‌مانده",
    "",
    "۱. ثبت تصمیم دربارهٔ تفاوت قرارداد Citation Coverage.",
    "",
    "۲. بررسی نسخهٔ نهایی نتایج NB22 و پایداری کیفیت؛ "
    "بدون بازاجرای بی‌دلیل آزمایش‌های قبلی.",
    "",
    "۳. تعیین پذیرش رسمی Release و سپس Handoff فاز سوم.",
    "",
    "۴. تکمیل ارجاع‌های شواهد پذیرش برای ردیف‌هایی "
    "که هنوز بررسی مستقل نشده‌اند.",
    "",
    "۵. ثبت تصمیم نهایی دربارهٔ چهار اختلاف دیگر نقشه‌ها.",
    "",
    "## ۹. محدودیت‌های گزارش",
    "",
    "- این سند ممیزی READ-ONLY است.",
    "- گزارش‌های نام‌دار، مجموعهٔ کامل شواهد پروژه نیستند.",
    "- صفر گزارش، اثبات نبود کار نیست.",
    "- نتایج Validation به‌تنهایی PASS نهایی نیستند.",
    "- هیچ Gate یا Release تازه‌ای پذیرفته نشده است.",
    "- تصمیم‌های رسمی تاریخی تغییر نکرده‌اند.",
])

P2AUD_C023_DELIVERABLE_MD = "\n".join(
    P2AUD_C023_LINES
)

P2AUD_C023_DELIVERABLE_SHA256 = hashlib.sha256(
    P2AUD_C023_DELIVERABLE_MD.encode("utf-8")
).hexdigest()

P2AUD_C023_DELIVERABLE_JSON = {
    "schema_version": "P2AUD_AS_BUILT_DELIVERABLE_V1",
    "status": "DRAFT_FOR_REVIEW",
    "document_sha256": P2AUD_C023_DELIVERABLE_SHA256,
    "source_roadmaps": {
        key: {
            "path": source["relative_path"],
            "sha256": source["sha256"],
        }
        for key, source in P2AUD_C022_SOURCES.items()
    },
    "as_built_notebooks": P2AUD_C022_AS_BUILT_MAP,
    "operational_package_crosswalk": (
        P2AUD_C021_PACKAGE_CROSSWALK
    ),
    "documentary_review_cases": (
        P2AUD_C022_REVIEW_CASES
    ),
    "historical_quality": P2AUD_C023_QUALITY,
    "unclassified_support_notebooks": (
        P2AUD_C015_UNCLASSIFIED_VARIANTS
    ),
    "formal_acceptance": False,
    "files_written": 0,
}

P2AUD_C023_REPORT = {
    "schema_version": "P2AUD_AS_BUILT_DELIVERY_SUMMARY_V1",
    "cell_id": "P2AUD-C023",
    "status": "AS_BUILT_DRAFT_DELIVERABLE_READY",
    "notebooks": len(P2AUD_C022_AS_BUILT_MAP),
    "execution_stages": len(
        P2AUD_C021_STAGE_DEFINITIONS
    ),
    "operational_packages": len(
        P2AUD_C021_PACKAGE_CROSSWALK
    ),
    "roadmap_cases": len(
        P2AUD_C022_REVIEW_CASES
    ),
    "document_sha256": P2AUD_C023_DELIVERABLE_SHA256,
    "document_lines": len(
        P2AUD_C023_DELIVERABLE_MD.splitlines()
    ),
    "documentary_reconciliation_complete": True,
    "formal_roadmap_decisions_approved": 0,
    "release_accepted": False,
    "files_written": 0,
    "api_calls": 0,
    "git_operations": 0,
    "next_step": "REVIEW_AND_APPROVE_AS_BUILT_DOCUMENT",
}

try:
    from IPython.display import display, Markdown
    display(Markdown(P2AUD_C023_DELIVERABLE_MD))
except ImportError:
    print(P2AUD_C023_DELIVERABLE_MD)

print("\n========== P2AUD-C023 SUMMARY ==========\n")
print(json.dumps(
    P2AUD_C023_REPORT,
    ensure_ascii=False,
    indent=2,
))
````

````python
# P2AUD-C024 — Register versioned As-Built deliverable
# LOCAL FILE OUTPUT | NO API | NO GIT | NO HISTORICAL OVERWRITE

import json
import copy
import hashlib
from pathlib import Path

assert P2AUD_C023_REPORT["status"] == (
    "AS_BUILT_DRAFT_DELIVERABLE_READY"
)
assert len(P2AUD_C022_AS_BUILT_MAP) == 24
assert len(P2AUD_C021_PACKAGE_CROSSWALK) == 14
assert len(P2AUD_C022_REVIEW_CASES) == 5

# Verify exact approved-for-review document snapshot.
md_content = P2AUD_C023_DELIVERABLE_MD
md_bytes = md_content.encode("utf-8")
md_sha = hashlib.sha256(md_bytes).hexdigest()

assert md_sha == P2AUD_C023_REPORT["document_sha256"]
assert len(md_content.splitlines()) == 236

# Preserve the As-Built draft and attach evidence registers.
P2AUD_C024_PACKAGE = copy.deepcopy(
    P2AUD_C023_DELIVERABLE_JSON
)

P2AUD_C024_PACKAGE["review_evidence"] = {
    "per_notebook": copy.deepcopy(
        P2AUD_C016_EVIDENCE_INDEX
    ),
    "historical_document_ledger": copy.deepcopy(
        P2AUD_C017_DOCUMENT_LEDGER
    ),
    "historical_checklist_ledger": copy.deepcopy(
        P2AUD_C017_CHECKLIST_LEDGER
    ),
    "nb22_bounded_report_artifacts": copy.deepcopy(
        P2AUD_C020_REPORT_ARTIFACTS
    ),
}

P2AUD_C024_PACKAGE["audit_closeout"] = {
    "documentary_reconciliation": "COMPLETE",
    "document_status": "DRAFT_READY_FOR_OWNER_REVIEW",
    "formal_roadmap_decisions_approved": 0,
    "all_notebook_acceptance_verified": False,
    "release_accepted": False,
    "phase3_handoff_accepted": False,
    "historical_decisions_modified": False,
    "remaining_work": [
        "Resolve frozen Citation Coverage requirement",
        "Verify current NB22 release-quality evidence",
        "Accept Phase 2 release through its formal gate",
        "Complete NB23 formal Phase 3 handoff",
        "Finish independent evidence review where pending",
    ],
}

json_bytes = (
    json.dumps(
        P2AUD_C024_PACKAGE,
        ensure_ascii=False,
        sort_keys=True,
        indent=2,
    ) + "\n"
).encode("utf-8")

json_sha = hashlib.sha256(json_bytes).hexdigest()

# Create new, versioned output files only.
repo = Path(P2AUD_REPO_ROOT).resolve()
out_dir = repo / "reports/phase2/audit/as_built"

assert out_dir.resolve().is_relative_to(repo)
out_dir.mkdir(parents=True, exist_ok=True)

prefix = "P2AUD.C024"

md_name = f"{prefix}.as_built.{md_sha[:16]}.md"
json_name = f"{prefix}.evidence_pack.{json_sha[:16]}.json"

def save_immutable(name, data):
    path = out_dir / name

    try:
        with path.open("xb") as handle:
            handle.write(data)
        status = "CREATED"
    except FileExistsError:
        if path.read_bytes() != data:
            raise RuntimeError(
                f"Existing file differs: {path}"
            )
        status = "REUSED_IDENTICAL"

    assert hashlib.sha256(
        path.read_bytes()
    ).hexdigest() == hashlib.sha256(data).hexdigest()

    return {
        "path": path.relative_to(repo).as_posix(),
        "sha256": hashlib.sha256(data).hexdigest(),
        "status": status,
        "bytes": len(data),
    }

md_receipt = save_immutable(md_name, md_bytes)
json_receipt = save_immutable(json_name, json_bytes)

receipt_document = {
    "schema_version": "P2AUD_C024_RECEIPT_V1",
    "status": "AS_BUILT_DRAFT_REGISTERED_FOR_REVIEW",
    "markdown": md_receipt,
    "evidence_json": json_receipt,
    "documentary_review_complete": True,
    "formal_release_acceptance": False,
    "formal_owner_approval": False,
    "historical_files_modified": False,
    "git_operations": 0,
    "api_calls": 0,
}

receipt_bytes = (
    json.dumps(
        receipt_document,
        ensure_ascii=False,
        sort_keys=True,
        indent=2,
    ) + "\n"
).encode("utf-8")

receipt_sha = hashlib.sha256(
    receipt_bytes
).hexdigest()

receipt = save_immutable(
    f"{prefix}.receipt.{receipt_sha[:16]}.json",
    receipt_bytes,
)

P2AUD_C024_REPORT = {
    "cell_id": "P2AUD-C024",
    "status": "AS_BUILT_DRAFT_REGISTERED_FOR_REVIEW",
    "notebooks": 24,
    "operational_packages": 14,
    "roadmap_review_cases": 5,
    "files": [
        md_receipt,
        json_receipt,
        receipt,
    ],
    "files_created_now": sum(
        item["status"] == "CREATED"
        for item in (md_receipt, json_receipt, receipt)
    ),
    "historical_files_modified": False,
    "formal_release_accepted": False,
    "owner_approval_pending": True,
    "next_step": "OWNER_REVIEW_AND_AUDIT_DOCUMENT_CLOSEOUT",
}

print(json.dumps(
    P2AUD_C024_REPORT,
    ensure_ascii=False,
    indent=2,
))
````
