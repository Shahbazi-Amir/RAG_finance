````python
# Code Cell 02

# =========================================================
# Helpers
# =========================================================

def sha256_file(
    path,
):

    digest = hashlib.sha256()

    with path.open(
        "rb",
    ) as file:

        for chunk in iter(
            lambda: file.read(
                1024 * 1024
            ),
            b"",
        ):

            digest.update(
                chunk
            )

    return digest.hexdigest()


def run_git(
    project_root,
    *args,
):

    result = subprocess.run(
        [
            "git",
            *args,
        ],
        cwd=project_root,
        check=True,
        capture_output=True,
        text=True,
    )

    return result.stdout.strip()


def load_json_file(
    path,
):

    with path.open(
        "r",
        encoding="utf-8",
    ) as file:

        return json.load(
            file
        )


def count_jsonl_records(
    path,
):

    count = 0

    with path.open(
        "r",
        encoding="utf-8",
    ) as file:

        for line_number, line in enumerate(
            file,
            start=1,
        ):

            if not line.strip():
                continue

            try:

                json.loads(
                    line
                )

            except json.JSONDecodeError as exc:

                raise AssertionError(
                    "Invalid JSONL record in "
                    f"{path} at line {line_number}"
                ) from exc

            count += 1

    return count


# =========================================================
# Discover Project Root
# =========================================================

current_working_directory = (
    Path.cwd().resolve()
)


project_root_candidates = []


for candidate in [
    current_working_directory,
    *current_working_directory.parents,
]:

    if (
        (
            candidate
            / ".git"
        ).exists()
        and (
            candidate
            / "notebooks"
            / "phase2"
        ).exists()
    ):

        project_root_candidates.append(
            candidate
        )


assert (
    len(
        project_root_candidates
    )
    == 1
), (
    "Could not resolve exactly one project root. "
    f"Candidates: {project_root_candidates}"
)


project_root = (
    project_root_candidates[
        0
    ]
)


git_toplevel = Path(
    run_git(
        project_root,
        "rev-parse",
        "--show-toplevel",
    )
).resolve()


assert (
    git_toplevel
    == project_root
), (
    "Filesystem project root and Git root differ."
)


# =========================================================
# Git read-only state
# =========================================================

git_branch = run_git(
    project_root,
    "rev-parse",
    "--abbrev-ref",
    "HEAD",
)


git_head = run_git(
    project_root,
    "rev-parse",
    "HEAD",
)


git_status_short = run_git(
    project_root,
    "status",
    "--short",
)


# =========================================================
# NB12j identity + historical 12-series inventory
#
# IMPORTANT:
# - Historical upstream notebooks are an explicit frozen set.
# - Any additional 12j copy/backup/repaired file is NOT treated
#   as a historical upstream notebook.
# - This makes Restart Kernel -> Run All safe even if another
#   12j copy exists beside the canonical notebook.
# =========================================================

phase2_notebooks_dir = (
    project_root
    / "notebooks"
    / "phase2"
)


canonical_notebook_path = (
    phase2_notebooks_dir
    / "12j_golden_dataset_v2_registry_driven_rebuild.ipynb"
)


twelve_series_notebooks = sorted(
    path
    for path in phase2_notebooks_dir.glob(
        "12*.ipynb"
    )
    if path.is_file()
)


assert (
    twelve_series_notebooks
), (
    "No 12-series Notebooks found."
)


nb12j_notebook_candidates = sorted(
    path
    for path in twelve_series_notebooks
    if path.name.startswith(
        "12j"
    )
)


assert (
    nb12j_notebook_candidates
), (
    "No NB12j notebook file was found in notebooks/phase2."
)


# Prefer the canonical production filename when it exists.
# Otherwise use the only discovered NB12j candidate.
if canonical_notebook_path.exists():

    notebook_path = (
        canonical_notebook_path
    )

else:

    assert (
        len(
            nb12j_notebook_candidates
        )
        == 1
    ), (
        "Canonical NB12j filename is missing and multiple "
        "NB12j variants exist. Rename the active notebook to:\n"
        f"{canonical_notebook_path}"
    )

    notebook_path = (
        nb12j_notebook_candidates[
            0
        ]
    )


historical_twelve_series_names = [
    "12_golden_dataset.ipynb",
    "12b_golden_dataset_continuation_compact.ipynb",
    "12c_human_review_for_author.ipynb",
    "12d_second_human_review.ipynb",
    "12e_golden_dataset_full_generation.ipynb",
    "12f_author_feedback_integration.ipynb",
    "12g_golden_dataset_quality_rebuild_lab.ipynb",
    "12g_golden_dataset_rebuild_clean.ipynb",
    "12h_pilot_and_expansion_audit.ipynb",
    "12i_claim_registry_and_knowledge_graph.ipynb",
]


historical_twelve_series_notebooks = [
    phase2_notebooks_dir
    / filename
    for filename in (
        historical_twelve_series_names
    )
]


missing_historical_twelve_series_notebooks = [
    path
    for path in (
        historical_twelve_series_notebooks
    )
    if not path.exists()
]


assert not (
    missing_historical_twelve_series_notebooks
), (
    "Missing expected historical 12-series notebooks:\n- "
    + "\n- ".join(
        str(
            path.relative_to(
                project_root
            )
        )
        for path in (
            missing_historical_twelve_series_notebooks
        )
    )
)


assert (
    len(
        historical_twelve_series_notebooks
    )
    == 10
)


historical_twelve_series_relative_paths = {
    str(
        path.relative_to(
            project_root
        )
    )
    for path in (
        historical_twelve_series_notebooks
    )
}


ignored_extra_twelve_series_notebooks = [
    path
    for path in (
        twelve_series_notebooks
    )
    if (
        str(
            path.relative_to(
                project_root
            )
        )
        not in historical_twelve_series_relative_paths
        and path.resolve()
        != notebook_path.resolve()
    )
]


historical_twelve_series_notebook_sha256 = {
    str(
        path.relative_to(
            project_root
        )
    ): sha256_file(
        path
    )
    for path in (
        historical_twelve_series_notebooks
    )
}


# Backward-compatible name used by later cells.
# It intentionally contains ONLY historical upstream notebooks.
twelve_series_notebook_sha256 = dict(
    historical_twelve_series_notebook_sha256
)


# =========================================================
# Final Registry paths from NB12i
# =========================================================

registry_root = (
    project_root
    / "artifacts"
    / "phase2"
    / "knowledge_registry"
)


registry_paths = {
    "contract": (
        registry_root
        / "claim_registry_contract_v1.json"
    ),

    "canonical_claims": (
        registry_root
        / "claims_deduplicated_v1.jsonl"
    ),

    "evidence_links": (
        registry_root
        / "claim_evidence_links_v1.jsonl"
    ),

    "claim_relations": (
        registry_root
        / "claim_relations_v1.jsonl"
    ),

    "knowledge_graph": (
        registry_root
        / "knowledge_graph_v1.json"
    ),

    "registry_report": (
        registry_root
        / "claim_registry_report_v1.json"
    ),

    "claim_id_mapping": (
        registry_root
        / "claim_id_mapping_v1.jsonl"
    ),
}


missing_registry_artifacts = [
    name
    for (
        name,
        path,
    ) in (
        registry_paths.items()
    )
    if not path.exists()
]


assert not missing_registry_artifacts, (
    "Missing final Registry artifacts:\n- "
    + "\n- ".join(
        missing_registry_artifacts
    )
)


# =========================================================
# Verify known frozen Registry hashes
# =========================================================

known_registry_sha256 = {
    "canonical_claims": (
        "8d1080d6519faa8ca5d4614256dbabf04426170b8f914f606eb3088d1f2c08bb"
    ),

    "evidence_links": (
        "1c47a493b2128b1fd489752378a3e5d4dc7e53b7379b03674abe37d9a77cc9ba"
    ),

    "claim_relations": (
        "b505a90036ea9eeec36ddb3e9ebd687f5503c4f8cfccf59f0fb0b684774d9009"
    ),

    "knowledge_graph": (
        "a408f12bac5470de46d700a3c8082e39899473984651b6ca1eeb377e8d80dd1a"
    ),

    "registry_report": (
        "f7951467ca49d1bac1e1c1efee9e93e756f24fbf6db02e500e35a6672b21d645"
    ),

    "claim_id_mapping": (
        "5b5e07134961212984e480e0574722ee2e12cef4aebf21922d7a99a8ab3c9ad2"
    ),
}


registry_sha256 = {
    name: sha256_file(
        path
    )
    for (
        name,
        path,
    ) in (
        registry_paths.items()
    )
}


for (
    artifact_name,
    expected_sha256,
) in (
    known_registry_sha256.items()
):

    actual_sha256 = (
        registry_sha256[
            artifact_name
        ]
    )

    assert (
        actual_sha256
        == expected_sha256
    ), (
        "Frozen Registry SHA mismatch:\n"
        f"Artifact: {artifact_name}\n"
        f"Expected: {expected_sha256}\n"
        f"Actual:   {actual_sha256}"
    )


# =========================================================
# Validate Final Registry Report
# =========================================================

registry_report = load_json_file(
    registry_paths[
        "registry_report"
    ]
)


assert (
    registry_report[
        "registry_status"
    ]
    == "PASS"
)


assert all(
    registry_report[
        "acceptance_checks"
    ].values()
)


registry_counts = (
    registry_report[
        "counts"
    ]
)


assert (
    registry_counts[
        "raw_claims"
    ]
    == 6237
)


assert (
    registry_counts[
        "validated_claims"
    ]
    == 4039
)


assert (
    registry_counts[
        "canonical_claims"
    ]
    == 4006
)


assert (
    registry_counts[
        "validated_registry_evidence_occurrences"
    ]
    == 4125
)


assert (
    registry_counts[
        "claim_relations_total"
    ]
    == 2516
)


assert (
    registry_counts[
        "graph_edge_eligible_claim_relations"
    ]
    == 2415
)


# =========================================================
# Discover legacy Golden / Core / Author Review
# =========================================================

search_roots = [
    project_root
    / "notebooks",

    project_root
    / "artifacts",

    project_root
    / "archive",

    project_root
    / "data",
]


search_roots = [
    path
    for path in search_roots
    if path.exists()
]


def find_unique_filename(
    filename,
):

    matches = []

    for search_root in (
        search_roots
    ):

        matches.extend(
            path.resolve()
            for path in (
                search_root.rglob(
                    filename
                )
            )
            if path.is_file()
        )

    matches = sorted(
        set(
            matches
        )
    )


    assert (
        len(
            matches
        )
        == 1
    ), (
        "Expected exactly one file named "
        f"{filename!r}, found {len(matches)}:\n"
        + "\n".join(
            str(
                path
            )
            for path in matches
        )
    )


    return matches[
        0
    ]


golden_v1_path = (
    find_unique_filename(
        "golden_dataset_v1.jsonl"
    )
)


golden_v1_report_path = (
    find_unique_filename(
        "golden_dataset_v1_final_report.json"
    )
)


core46_path = (
    find_unique_filename(
        "golden_dataset_quote_grounded_final_adjudicated_experiment.jsonl"
    )
)


author_review_path = (
    find_unique_filename(
        "author_review_28_pilots.md"
    )
)


# =========================================================
# Validate legacy JSONL artifacts structurally
# =========================================================

golden_v1_record_count = (
    count_jsonl_records(
        golden_v1_path
    )
)


core46_record_count = (
    count_jsonl_records(
        core46_path
    )
)


golden_v1_sha256 = sha256_file(
    golden_v1_path
)


golden_v1_report_sha256 = sha256_file(
    golden_v1_report_path
)


core46_sha256 = sha256_file(
    core46_path
)


author_review_sha256 = sha256_file(
    author_review_path
)


# =========================================================
# Preflight Gate
# =========================================================

preflight_status = (
    "PASS"
)


next_stage_readiness = (
    "READY_FOR_12_SERIES_CONSOLIDATION_CONTRACT"
)


# =========================================================
# Result
# =========================================================

print(
    "=== NB12j REPOSITORY / ARTIFACT PREFLIGHT ==="
)

print()

print(
    "Project Root:",
    project_root,
)

print(
    "Current Working Directory:",
    current_working_directory,
)

print(
    "Git Branch:",
    git_branch,
)

print(
    "Git HEAD:",
    git_head,
)

print(
    "Git working tree dirty:",
    bool(
        git_status_short
    ),
)


print()

print(
    "Canonical NB12j path:",
    notebook_path.relative_to(
        project_root
    ),
)

print(
    "NB12j variants discovered:",
    len(
        nb12j_notebook_candidates
    ),
)

for path in (
    nb12j_notebook_candidates
):

    print(
        "  -",
        path.relative_to(
            project_root
        ),
    )


print()

print(
    "Historical 12-series Notebooks:",
    len(
        historical_twelve_series_notebooks
    ),
)


for path in (
    historical_twelve_series_notebooks
):

    relative_path = str(
        path.relative_to(
            project_root
        )
    )

    print(
        "  -",
        relative_path,
        historical_twelve_series_notebook_sha256[
            relative_path
        ],
    )


print()

print(
    "Ignored extra 12-series notebook variants:",
    len(
        ignored_extra_twelve_series_notebooks
    ),
)

for path in (
    ignored_extra_twelve_series_notebooks
):

    print(
        "  -",
        path.relative_to(
            project_root
        ),
    )


print()

print(
    "Final Registry status:",
    registry_report[
        "registry_status"
    ],
)

print(
    "Final Registry acceptance checks:",
    sum(
        registry_report[
            "acceptance_checks"
        ].values()
    ),
    "/",
    len(
        registry_report[
            "acceptance_checks"
        ]
    ),
    "PASS",
)


print()

print(
    "Canonical Claims:",
    registry_counts[
        "canonical_claims"
    ],
)

print(
    "Validated Evidence occurrences:",
    registry_counts[
        "validated_registry_evidence_occurrences"
    ],
)

print(
    "Claim Relations:",
    registry_counts[
        "claim_relations_total"
    ],
)

print(
    "Graph-edge eligible Relations:",
    registry_counts[
        "graph_edge_eligible_claim_relations"
    ],
)


print()

print(
    "Golden v1:",
    golden_v1_path.relative_to(
        project_root
    ),
)

print(
    "Golden v1 records:",
    golden_v1_record_count,
)

print(
    "Golden v1 SHA-256:",
    golden_v1_sha256,
)


print()

print(
    "Golden v1 Final Report:",
    golden_v1_report_path.relative_to(
        project_root
    ),
)

print(
    "Golden v1 Final Report SHA-256:",
    golden_v1_report_sha256,
)


print()

print(
    "Core-46:",
    core46_path.relative_to(
        project_root
    ),
)

print(
    "Core-46 records:",
    core46_record_count,
)

print(
    "Core-46 SHA-256:",
    core46_sha256,
)


print()

print(
    "Author Review:",
    author_review_path.relative_to(
        project_root
    ),
)

print(
    "Author Review SHA-256:",
    author_review_sha256,
)


print()

print(
    "Files modified by this Cell:",
    0,
)

print(
    "API calls performed:",
    0,
)


print()

print(
    "Next stage:",
    "12_SERIES_CONSOLIDATION_CONTRACT",
)

print(
    "Readiness:",
    next_stage_readiness,
)


print()

print(
    "CODE CELL 02 GATE:",
    preflight_status,
)
````

````python
# Code Cell 03

assert (
    preflight_status
    == "PASS"
), (
    "Code Cell 02 must PASS before "
    "freezing the NB12j Contract."
)

assert (
    next_stage_readiness
    == "READY_FOR_12_SERIES_CONSOLIDATION_CONTRACT"
)


# =========================================================
# Output namespace
# =========================================================

golden_v2_artifact_root = (
    project_root
    / "artifacts"
    / "phase2"
    / "golden_dataset_v2"
)


golden_v2_artifact_root.mkdir(
    parents=True,
    exist_ok=True,
)


golden_v2_contract_path = (
    golden_v2_artifact_root
    / "golden_dataset_v2_registry_driven_contract_v1.json"
)


# =========================================================
# Frozen publishing helper
#
# Existing file:
#   identical -> REUSED_EXISTING_IDENTICAL
#   different -> FAIL
#
# No destructive overwrite.
# =========================================================

def publish_frozen_text(
    path,
    text,
):

    encoded = text.encode(
        "utf-8"
    )


    if path.exists():

        existing = path.read_bytes()

        assert (
            existing
            == encoded
        ), (
            "Frozen artifact already exists with "
            "different content:\n"
            f"{path}"
        )

        return (
            "REUSED_EXISTING_IDENTICAL"
        )


    path.write_bytes(
        encoded
    )

    return (
        "CREATED_NEW"
    )


# =========================================================
# Source manifests from actual Preflight
# =========================================================

current_notebook_relative_path = (
    "notebooks/phase2/"
    "12j_golden_dataset_v2_registry_driven_rebuild.ipynb"
)


# Freeze hashes only for the explicit historical upstream set.
# Any active/backup/repaired NB12j file is intentionally excluded.
twelve_series_source_manifest = {
    relative_path: {
        "sha256": sha256_value,
        "execution_policy": (
            "DO_NOT_RERUN_FOR_CONSOLIDATION"
        ),
        "mutation_policy": (
            "READ_ONLY"
        ),
    }
    for (
        relative_path,
        sha256_value,
    ) in sorted(
        historical_twelve_series_notebook_sha256.items()
    )
}


assert (
    len(
        twelve_series_source_manifest
    )
    == 10
), (
    "Historical 12-series source manifest must contain "
    "exactly the 10 validated upstream notebooks."
)


registry_source_manifest = {
    name: {
        "path": str(
            path.relative_to(
                project_root
            )
        ),
        "sha256": (
            registry_sha256[
                name
            ]
        ),
    }
    for (
        name,
        path,
    ) in sorted(
        registry_paths.items()
    )
}


legacy_golden_source_manifest = {
    "golden_v1": {
        "path": str(
            golden_v1_path.relative_to(
                project_root
            )
        ),
        "sha256": (
            golden_v1_sha256
        ),
        "records": (
            golden_v1_record_count
        ),
        "role": (
            "HISTORICAL_BASELINE"
        ),
        "mutation_policy": (
            "IMMUTABLE"
        ),
    },

    "golden_v1_final_report": {
        "path": str(
            golden_v1_report_path.relative_to(
                project_root
            )
        ),
        "sha256": (
            golden_v1_report_sha256
        ),
        "role": (
            "HISTORICAL_BASELINE_REPORT"
        ),
        "mutation_policy": (
            "IMMUTABLE"
        ),
    },

    "core46": {
        "path": str(
            core46_path.relative_to(
                project_root
            )
        ),
        "sha256": (
            core46_sha256
        ),
        "records": (
            core46_record_count
        ),
        "role": (
            "QUALITY_AND_REGRESSION_ANCHOR"
        ),
        "mutation_policy": (
            "IMMUTABLE"
        ),
    },

    "author_review": {
        "path": str(
            author_review_path.relative_to(
                project_root
            )
        ),
        "sha256": (
            author_review_sha256
        ),
        "role": (
            "AUTHOR_REVIEW_SOURCE"
        ),
        "review_status": (
            "TO_BE_DISCOVERED_FROM_SOURCE"
        ),
        "mutation_policy": (
            "IMMUTABLE"
        ),
    },
}


# =========================================================
# Planned artifact family
#
# These are NEW NB12j-owned outputs.
# Previous 12-series artifacts remain untouched.
# =========================================================

golden_v2_artifact_family = {
    "contract": (
        "golden_dataset_v2_registry_driven_contract_v1.json"
    ),

    "twelve_series_manifest": (
        "twelve_series_consolidated_manifest_v1.json"
    ),

    "claim_eligibility_manifest": (
        "golden_v2_claim_eligibility_manifest_v1.jsonl"
    ),

    "seed_pool": (
        "golden_v2_seed_pool_v1.jsonl"
    ),

    "dataset_candidate": (
        "golden_dataset_v2_registry_driven_candidate_v1.jsonl"
    ),

    "candidate_report": (
        "golden_dataset_v2_registry_driven_candidate_report_v1.json"
    ),

    "final_acceptance_report": (
        "golden_dataset_v2_final_acceptance_report_v1.json"
    ),
}


# =========================================================
# Contract
# =========================================================

golden_v2_contract = {
    "contract_version": (
        "golden_dataset_v2_registry_driven_contract_v1"
    ),

    "notebook": (
        "notebooks/phase2/"
        "12j_golden_dataset_v2_registry_driven_rebuild.ipynb"
    ),

    "purpose": (
        "Consolidate the validated outputs of the "
        "12-series workflow and build a broad, "
        "registry-driven, evidence-grounded Golden "
        "Dataset v2 without modifying frozen upstream "
        "artifacts."
    ),

    "pipeline_position": [
        "SOURCE_CORPUS",
        "CLAIM_REGISTRY_AND_KNOWLEDGE_GRAPH",
        "GOLDEN_DATASET_V2",
        "BENCHMARK",
    ],

    "source_state": {
        "git_branch": (
            git_branch
        ),

        "git_head_at_preflight": (
            git_head
        ),

        "historical_twelve_series_notebook_count": (
            len(
                historical_twelve_series_notebooks
            )
        ),

        "nb12j_variant_count_at_preflight": (
            len(
                nb12j_notebook_candidates
            )
        ),

        "registry_status": (
            registry_report[
                "registry_status"
            ]
        ),

        "registry_acceptance_checks_passed": (
            sum(
                registry_report[
                    "acceptance_checks"
                ].values()
            )
        ),

        "registry_acceptance_checks_total": (
            len(
                registry_report[
                    "acceptance_checks"
                ]
            )
        ),

        "canonical_claims": (
            registry_counts[
                "canonical_claims"
            ]
        ),

        "validated_evidence_occurrences": (
            registry_counts[
                "validated_registry_evidence_occurrences"
            ]
        ),

        "claim_relations": (
            registry_counts[
                "claim_relations_total"
            ]
        ),
    },

    "source_manifests": {
        "twelve_series_notebooks": (
            twelve_series_source_manifest
        ),

        "claim_registry": (
            registry_source_manifest
        ),

        "legacy_golden_and_review": (
            legacy_golden_source_manifest
        ),
    },

    "consolidation_policy": {
        "rerun_old_12_series_notebooks": False,

        "copy_old_notebook_code_into_12j": False,

        "modify_old_12_series_notebooks": False,

        "modify_old_artifacts": False,

        "status_must_be_evidence_based": True,

        "unknown_status_when_not_provable": True,

        "preserve_source_hashes": True,

        "consolidation_unit": (
            "NOTEBOOK_AND_ARTIFACT_PROVENANCE"
        ),
    },

    "registry_consumption_policy": {
        "primary_seed_population_artifact": (
            "claims_deduplicated_v1.jsonl"
        ),

        "canonical_claim_population": (
            registry_counts[
                "canonical_claims"
            ]
        ),

        "raw_claims_are_direct_seed_population": False,

        "claim_id_required": True,

        "evidence_required": True,

        "source_provenance_required": True,

        "automatic_semantic_merge_allowed": False,

        "automatic_subsumption_merge_allowed": False,

        "original_source_remains_final_evidence": True,

        "retrieval_chunks_define_golden_truth": False,
    },

    "golden_seed_policy": {
        "seed_unit": (
            "INDEPENDENT_EVIDENCE_GROUNDED_CANONICAL_CLAIM"
        ),

        "paraphrase_counts_as_new_seed": False,

        "question_variant_counts_as_new_seed": False,

        "seed_requires_canonical_claim_id": True,

        "seed_requires_validated_evidence": True,

        "seed_requires_traceable_source": True,

        "seed_count_target": (
            "TO_BE_DERIVED_FROM_ACTUAL_ELIGIBILITY_AND_COVERAGE"
        ),

        "do_not_invent_fixed_seed_count": True,
    },

    "legacy_golden_policy": {
        "golden_v1_role": (
            "HISTORICAL_BASELINE"
        ),

        "golden_v1_mutable": False,

        "core46_role": (
            "QUALITY_AND_REGRESSION_ANCHOR"
        ),

        "core46_mutable": False,

        "core46_is_full_knowledge_universe": False,
    },

    "author_review_policy": {
        "status_source": (
            "author_review_28_pilots.md"
        ),

        "status_must_be_discovered": True,

        "assume_feedback_complete": False,

        "engineering_candidate_allowed_while_pending": True,

        "author_reviewed_final_freeze_requires_real_feedback": True,

        "fabricate_author_feedback": False,
    },

    "golden_generation_policy": {
        "evidence_first": True,

        "answer_must_be_grounded": True,

        "question_must_be_supported_by_claim_and_evidence": True,

        "preserve_claim_id": True,

        "preserve_evidence_id": True,

        "preserve_source_lineage": True,

        "control_seed_redundancy": True,

        "control_question_redundancy": True,

        "measure_registry_coverage": True,

        "measure_source_coverage": True,

        "measure_claim_type_coverage": True,

        "measure_authority_coverage": True,

        "negative_or_challenge_cases_must_not_break_grounding": True,
    },

    "artifact_namespace": str(
        golden_v2_artifact_root.relative_to(
            project_root
        )
    ),

    "artifact_family": (
        golden_v2_artifact_family
    ),

    "write_policy": {
        "versioned_outputs": True,

        "destructive_overwrite": False,

        "reuse_existing_only_if_byte_identical": True,

        "git_write_without_explicit_permission": False,
    },

    "final_freeze_policy": {
        "engineering_acceptance_required": True,

        "restart_kernel_run_all_required": True,

        "final_notebook_health_required": True,

        "author_reviewed_label_requires_completed_author_feedback": True,
    },

    "current_state": (
        "CONTRACT_FROZEN_PENDING_12_SERIES_EVIDENCE_DISCOVERY"
    ),
}


# =========================================================
# Deterministic serialization / rerun-safe freeze
# =========================================================

proposed_golden_v2_contract_text = (
    json.dumps(
        golden_v2_contract,
        ensure_ascii=False,
        indent=2,
        sort_keys=True,
    )
    + "\n"
)


def validate_existing_golden_v2_contract(
    existing_contract,
):

    assert (
        existing_contract.get(
            "contract_version"
        )
        == "golden_dataset_v2_registry_driven_contract_v1"
    )

    assert (
        existing_contract.get(
            "notebook"
        )
        == current_notebook_relative_path
    )

    source_state = existing_contract.get(
        "source_state",
        {},
    )

    assert (
        source_state.get(
            "canonical_claims"
        )
        == 4006
    )

    assert (
        source_state.get(
            "registry_status"
        )
        == "PASS"
    )

    assert (
        existing_contract.get(
            "artifact_family"
        )
        == golden_v2_artifact_family
    )

    existing_source_manifests = existing_contract.get(
        "source_manifests",
        {},
    )

    assert (
        existing_source_manifests.get(
            "claim_registry"
        )
        == registry_source_manifest
    ), (
        "Existing frozen contract no longer matches the "
        "frozen Claim Registry inputs."
    )

    assert (
        existing_source_manifests.get(
            "legacy_golden_and_review"
        )
        == legacy_golden_source_manifest
    ), (
        "Existing frozen contract no longer matches the "
        "legacy Golden / review inputs."
    )

    existing_twelve_series = existing_source_manifests.get(
        "twelve_series_notebooks",
        {},
    )

    # Historical upstream notebooks must still match exactly.
    # An older valid v1 contract may also contain a hash for NB12j
    # itself; that self-referential entry is intentionally ignored
    # during rerun validation because NB12j is the active notebook.
    for (
        relative_path,
        expected_entry,
    ) in twelve_series_source_manifest.items():

        assert (
            relative_path
            in existing_twelve_series
        ), (
            "Existing contract is missing historical notebook: "
            f"{relative_path}"
        )

        assert (
            existing_twelve_series[
                relative_path
            ][
                "sha256"
            ]
            == expected_entry[
                "sha256"
            ]
        ), (
            "Historical notebook hash changed after contract freeze: "
            f"{relative_path}"
        )

    assert (
        existing_contract[
            "golden_seed_policy"
        ][
            "paraphrase_counts_as_new_seed"
        ]
        is False
    )

    assert (
        existing_contract[
            "author_review_policy"
        ][
            "assume_feedback_complete"
        ]
        is False
    )


if golden_v2_contract_path.exists():

    disk_golden_v2_contract = load_json_file(
        golden_v2_contract_path
    )

    validate_existing_golden_v2_contract(
        disk_golden_v2_contract
    )

    golden_v2_contract = (
        disk_golden_v2_contract
    )

    golden_v2_contract_text = (
        golden_v2_contract_path.read_text(
            encoding="utf-8"
        )
    )

    golden_v2_contract_sha256 = sha256_file(
        golden_v2_contract_path
    )

    golden_v2_contract_write_status = (
        "REUSED_EXISTING_VALIDATED"
    )

else:

    golden_v2_contract_text = (
        proposed_golden_v2_contract_text
    )

    golden_v2_contract_sha256 = (
        hashlib.sha256(
            golden_v2_contract_text.encode(
                "utf-8"
            )
        ).hexdigest()
    )

    golden_v2_contract_write_status = (
        publish_frozen_text(
            golden_v2_contract_path,
            golden_v2_contract_text,
        )
    )

    disk_golden_v2_contract = load_json_file(
        golden_v2_contract_path
    )

    assert (
        disk_golden_v2_contract
        == golden_v2_contract
    )

    assert (
        sha256_file(
            golden_v2_contract_path
        )
        == golden_v2_contract_sha256
    )


# =========================================================
# Independent disk revalidation
# =========================================================

assert (
    sha256_file(
        golden_v2_contract_path
    )
    == golden_v2_contract_sha256
)

assert (
    golden_v2_contract[
        "source_state"
    ][
        "canonical_claims"
    ]
    == 4006
)

assert (
    golden_v2_contract[
        "source_state"
    ][
        "registry_status"
    ]
    == "PASS"
)

assert (
    golden_v2_contract[
        "golden_seed_policy"
    ][
        "paraphrase_counts_as_new_seed"
    ]
    is False
)

assert (
    golden_v2_contract[
        "author_review_policy"
    ][
        "assume_feedback_complete"
    ]
    is False
)


# =========================================================
# Gate
# =========================================================

consolidation_contract_status = (
    "PASS"
)


twelve_series_evidence_discovery_readiness = (
    "READY_FOR_12_SERIES_EVIDENCE_DISCOVERY"
)


# =========================================================
# Result
# =========================================================

print(
    "=== NB12j CONSOLIDATION / GOLDEN V2 CONTRACT ==="
)

print()

print(
    "Historical 12-series source Notebooks hashed:",
    len(
        twelve_series_source_manifest
    ),
)

print(
    "Historical 12-series upstream Notebooks:",
    len(
        historical_twelve_series_notebooks
    ),
)

print(
    "NB12j variants present:",
    len(
        nb12j_notebook_candidates
    ),
)

print(
    "Canonical Registry Claims:",
    registry_counts[
        "canonical_claims"
    ],
)

print(
    "Golden v1 baseline records:",
    golden_v1_record_count,
)

print(
    "Core-46 anchor records:",
    core46_record_count,
)


print()

print(
    "Output namespace:",
    golden_v2_artifact_root.relative_to(
        project_root
    ),
)

print(
    "Planned NB12j artifact family:",
    len(
        golden_v2_artifact_family
    ),
)


for (
    role,
    filename,
) in sorted(
    golden_v2_artifact_family.items()
):

    print(
        "  -",
        role,
        "->",
        filename,
    )


print()

print(
    "Old 12-series notebooks mutable:",
    False,
)

print(
    "Old Golden artifacts mutable:",
    False,
)

print(
    "Raw Claims used directly as Seeds:",
    False,
)

print(
    "Paraphrase counts as new Seed:",
    False,
)

print(
    "Author Feedback assumed complete:",
    False,
)

print(
    "Git writes performed:",
    0,
)

print(
    "API calls performed:",
    0,
)


print()

print(
    "Contract:",
    golden_v2_contract_path.relative_to(
        project_root
    ),
)

print(
    "Contract SHA-256:",
    golden_v2_contract_sha256,
)

print(
    "Write status:",
    golden_v2_contract_write_status,
)


print()

print(
    "Next stage:",
    "12_SERIES_EVIDENCE_DISCOVERY",
)

print(
    "Readiness:",
    twelve_series_evidence_discovery_readiness,
)


print()

print(
    "CODE CELL 03 GATE:",
    consolidation_contract_status,
)
````

````python
# Code Cell 04

assert (
    consolidation_contract_status
    == "PASS"
), (
    "Code Cell 03 must PASS before "
    "12-series Evidence Discovery."
)

assert (
    twelve_series_evidence_discovery_readiness
    == "READY_FOR_12_SERIES_EVIDENCE_DISCOVERY"
)


# =========================================================
# Helpers for .ipynb inspection
# =========================================================

def cell_source_text(
    cell,
):

    source = cell.get(
        "source",
        "",
    )

    if isinstance(
        source,
        list,
    ):

        source = "".join(
            source
        )

    return source


def output_text(
    output,
):

    output_type = output.get(
        "output_type"
    )


    if (
        output_type
        == "stream"
    ):

        text = output.get(
            "text",
            "",
        )

        if isinstance(
            text,
            list,
        ):

            text = "".join(
                text
            )

        return text


    if (
        output_type
        in {
            "execute_result",
            "display_data",
        }
    ):

        data = output.get(
            "data",
            {},
        )

        text = data.get(
            "text/plain",
            "",
        )

        if isinstance(
            text,
            list,
        ):

            text = "".join(
                text
            )

        return text


    if (
        output_type
        == "error"
    ):

        return (
            "ERROR "
            + str(
                output.get(
                    "ename",
                    "",
                )
            )
            + ": "
            + str(
                output.get(
                    "evalue",
                    "",
                )
            )
        )


    return ""


def first_markdown_heading(
    notebook_json,
):

    for cell in notebook_json.get(
        "cells",
        []
    ):

        if (
            cell.get(
                "cell_type"
            )
            != "markdown"
        ):

            continue


        source = cell_source_text(
            cell
        )


        # Markdown heading
        match = re.search(
            r"(?m)^\s*#{1,3}\s+(.+?)\s*$",
            source,
        )

        if match:

            return (
                match.group(
                    1
                ).strip()
            )


        # HTML heading
        match = re.search(
            r"<h[1-3][^>]*>(.*?)</h[1-3]>",
            source,
            flags=re.IGNORECASE
            | re.DOTALL,
        )

        if match:

            heading = re.sub(
                r"<[^>]+>",
                "",
                match.group(
                    1
                ),
            )

            return (
                heading.strip()
            )


    return None


def extract_signal_lines(
    text,
):

    signal_patterns = [
        r"\bGATE\b",
        r"\bPASS\b",
        r"\bFAIL\b",
        r"FINAL .*STATUS",
        r"FINAL .*HEALTH",
        r"NOTEBOOK HEALTH",
        r"NOTEBOOK OBJECTIVE",
        r"GOAL_ACHIEVED",
        r"READY_FOR_",
        r"BLOCKED",
        r"PENDING",
        r"FROZEN",
    ]


    combined_pattern = re.compile(
        "|".join(
            signal_patterns
        ),
        flags=re.IGNORECASE,
    )


    signals = []


    for raw_line in text.splitlines():

        line = raw_line.strip()


        if (
            line
            and combined_pattern.search(
                line
            )
        ):

            signals.append(
                line
            )


    # Preserve order, remove duplicates.
    deduplicated = []

    seen = set()


    for line in signals:

        if line in seen:
            continue

        seen.add(
            line
        )

        deduplicated.append(
            line
        )


    return deduplicated


# =========================================================
# Inspect PRIOR 12-series Notebooks
# Current NB12j is excluded from historical discovery.
# =========================================================

# Use the explicit historical upstream set from Code Cell 02.
# Do not derive history by subtracting one active filename from
# a glob, because local backup/repaired NB12j copies may exist.
prior_twelve_series_notebooks = list(
    historical_twelve_series_notebooks
)


assert (
    len(
        prior_twelve_series_notebooks
    )
    == 10
), (
    "Expected the validated historical 12-series set "
    "to contain exactly 10 notebooks."
)


twelve_series_evidence = []


for path in (
    prior_twelve_series_notebooks
):

    notebook_json = (
        load_json_file(
            path
        )
    )


    cells = notebook_json.get(
        "cells",
        []
    )


    code_cells = [
        cell
        for cell in cells
        if (
            cell.get(
                "cell_type"
            )
            == "code"
        )
    ]


    markdown_cells = [
        cell
        for cell in cells
        if (
            cell.get(
                "cell_type"
            )
            == "markdown"
        )
    ]


    executed_code_cells = [
        cell
        for cell in code_cells
        if (
            cell.get(
                "execution_count"
            )
            is not None
        )
    ]


    unexecuted_code_cells = [
        cell
        for cell in code_cells
        if (
            cell.get(
                "execution_count"
            )
            is None
        )
    ]


    error_outputs = []


    stored_output_fragments = []


    for (
        cell_index,
        cell,
    ) in enumerate(
        code_cells,
        start=1,
    ):

        for output in cell.get(
            "outputs",
            []
        ):

            text = output_text(
                output
            )


            if text:

                stored_output_fragments.append(
                    text
                )


            if (
                output.get(
                    "output_type"
                )
                == "error"
            ):

                error_outputs.append({
                    "code_cell_position": (
                        cell_index
                    ),

                    "ename": output.get(
                        "ename"
                    ),

                    "evalue": output.get(
                        "evalue"
                    ),
                })


    stored_outputs_text = "\n".join(
        stored_output_fragments
    )


    signal_lines = extract_signal_lines(
        stored_outputs_text
    )


    execution_counts = [
        cell.get(
            "execution_count"
        )
        for cell in (
            executed_code_cells
        )
        if isinstance(
            cell.get(
                "execution_count"
            ),
            int,
        )
    ]


    last_execution_count = (
        max(
            execution_counts
        )
        if execution_counts
        else None
    )


    twelve_series_evidence.append({
        "path": str(
            path.relative_to(
                project_root
            )
        ),

        "sha256": (
            twelve_series_notebook_sha256[
                str(
                    path.relative_to(
                        project_root
                    )
                )
            ]
        ),

        "first_heading": (
            first_markdown_heading(
                notebook_json
            )
        ),

        "markdown_cells": (
            len(
                markdown_cells
            )
        ),

        "code_cells": (
            len(
                code_cells
            )
        ),

        "executed_code_cells": (
            len(
                executed_code_cells
            )
        ),

        "unexecuted_code_cells": (
            len(
                unexecuted_code_cells
            )
        ),

        "stored_error_outputs": (
            len(
                error_outputs
            )
        ),

        "error_evidence": (
            error_outputs
        ),

        "last_execution_count": (
            last_execution_count
        ),

        "signal_lines": (
            signal_lines
        ),
    })


# =========================================================
# Author Review evidence discovery
# =========================================================

author_review_text = author_review_path.read_text(
    encoding="utf-8"
)


author_review_lines = (
    author_review_text.splitlines()
)


author_review_marker_lines = [
    line.strip()
    for line in author_review_lines
    if (
        "نظر مؤلف"
        in line
    )
]


review_case_markers = [
    line.strip()
    for line in author_review_lines
    if re.search(
        r"(?i)\breview\s*case\b",
        line,
    )
]


# Conservative discovery:
# collect text occurring on the SAME line after "نظر مؤلف".
# We do not infer completion from surrounding structure yet.
author_review_inline_values = []


for line in (
    author_review_marker_lines
):

    marker_position = line.find(
        "نظر مؤلف"
    )


    suffix = line[
        marker_position
        + len(
            "نظر مؤلف"
        ):
    ].strip()


    suffix = suffix.lstrip(
        ":：-*# "
    ).strip()


    if suffix:

        author_review_inline_values.append(
            suffix
        )


# =========================================================
# Golden v1 report structural discovery
# =========================================================

golden_v1_report = load_json_file(
    golden_v1_report_path
)


golden_v1_report_keys = sorted(
    golden_v1_report.keys()
)


# =========================================================
# Golden v1 first-record schema
# =========================================================

def first_jsonl_record(
    path,
):

    with path.open(
        "r",
        encoding="utf-8",
    ) as file:

        for line in file:

            if line.strip():

                return json.loads(
                    line
                )


    raise AssertionError(
        f"No JSONL records found: {path}"
    )


golden_v1_first_record = (
    first_jsonl_record(
        golden_v1_path
    )
)


core46_first_record = (
    first_jsonl_record(
        core46_path
    )
)


golden_v1_record_keys = sorted(
    golden_v1_first_record.keys()
)


core46_record_keys = sorted(
    core46_first_record.keys()
)


# =========================================================
# No writes in this Step
# =========================================================

twelve_series_evidence_discovery_status = (
    "PASS"
)


twelve_series_manifest_readiness = (
    "READY_FOR_12_SERIES_CONSOLIDATED_MANIFEST"
)


# =========================================================
# Result
# =========================================================

print(
    "=== NB12j 12-SERIES EVIDENCE DISCOVERY ==="
)

print()

print(
    "Prior 12-series Notebooks inspected:",
    len(
        twelve_series_evidence
    ),
)


for evidence in (
    twelve_series_evidence
):

    print()
    print(
        evidence[
            "path"
        ]
    )

    print(
        "  Heading:",
        evidence[
            "first_heading"
        ],
    )

    print(
        "  Markdown / Code:",
        evidence[
            "markdown_cells"
        ],
        "/",
        evidence[
            "code_cells"
        ],
    )

    print(
        "  Executed / Unexecuted Code:",
        evidence[
            "executed_code_cells"
        ],
        "/",
        evidence[
            "unexecuted_code_cells"
        ],
    )

    print(
        "  Stored Error outputs:",
        evidence[
            "stored_error_outputs"
        ],
    )

    print(
        "  Last execution count:",
        evidence[
            "last_execution_count"
        ],
    )

    print(
        "  Explicit signal lines:",
        len(
            evidence[
                "signal_lines"
            ]
        ),
    )


    for line in (
        evidence[
            "signal_lines"
        ][
            -12:
        ]
    ):

        print(
            "    ",
            line,
        )


print()
print(
    "=== AUTHOR REVIEW DISCOVERY ==="
)

print(
    "Review Case markers:",
    len(
        review_case_markers
    ),
)

print(
    "'نظر مؤلف' markers:",
    len(
        author_review_marker_lines
    ),
)

print(
    "Non-empty inline values after 'نظر مؤلف':",
    len(
        author_review_inline_values
    ),
)


if author_review_inline_values:

    print(
        "Inline value samples:"
    )

    for value in (
        author_review_inline_values[
            :10
        ]
    ):

        print(
            "  -",
            value,
        )


print()
print(
    "=== LEGACY GOLDEN STRUCTURE ==="
)

print(
    "Golden v1 records:",
    golden_v1_record_count,
)

print(
    "Golden v1 first-record keys:",
    golden_v1_record_keys,
)

print(
    "Golden v1 report top-level keys:",
    golden_v1_report_keys,
)


print()

print(
    "Core-46 records:",
    core46_record_count,
)

print(
    "Core-46 first-record keys:",
    core46_record_keys,
)


print()

print(
    "Files modified by this Cell:",
    0,
)

print(
    "API calls performed:",
    0,
)


print()

print(
    "Next stage:",
    "12_SERIES_CONSOLIDATED_MANIFEST",
)

print(
    "Readiness:",
    twelve_series_manifest_readiness,
)


print()

print(
    "CODE CELL 04 GATE:",
    twelve_series_evidence_discovery_status,
)
````

````python
# Code Cell 06

assert (
    twelve_series_manifest_status
    == "PASS"
), (
    "Code Cell 05 must PASS before "
    "Registry Claim Eligibility Profile."
)

assert (
    claim_eligibility_profile_readiness
    == "READY_FOR_REGISTRY_CLAIM_ELIGIBILITY_PROFILE"
)


# =========================================================
# Helpers
# =========================================================

def load_jsonl_records(
    path,
):

    records = []

    with path.open(
        "r",
        encoding="utf-8",
    ) as file:

        for line_number, line in enumerate(
            file,
            start=1,
        ):

            if not line.strip():
                continue

            try:

                record = json.loads(
                    line
                )

            except json.JSONDecodeError as exc:

                raise AssertionError(
                    "Invalid JSONL record:\n"
                    f"{path}\n"
                    f"Line: {line_number}"
                ) from exc

            records.append(
                record
            )

    return records


def numeric_profile(
    values,
):

    assert values

    ordered = sorted(
        values
    )

    count = len(
        ordered
    )


    def percentile(
        fraction,
    ):

        if count == 1:
            return ordered[0]

        position = (
            count - 1
        ) * fraction

        lower_index = int(
            position
        )

        upper_index = min(
            lower_index + 1,
            count - 1,
        )

        weight = (
            position
            - lower_index
        )

        return (
            ordered[
                lower_index
            ]
            * (
                1
                - weight
            )
            + ordered[
                upper_index
            ]
            * weight
        )


    return {
        "count": (
            count
        ),

        "min": (
            ordered[0]
        ),

        "p50": (
            percentile(
                0.50
            )
        ),

        "p90": (
            percentile(
                0.90
            )
        ),

        "p95": (
            percentile(
                0.95
            )
        ),

        "p99": (
            percentile(
                0.99
            )
        ),

        "max": (
            ordered[-1]
        ),

        "mean": (
            sum(
                ordered
            )
            / count
        ),
    }


def sorted_counter_dict(
    counter,
):

    return dict(
        sorted(
            counter.items(),
            key=lambda item: (
                str(
                    item[0]
                )
            ),
        )
    )


def is_nonempty_value(
    value,
):

    if value is None:
        return False

    if isinstance(
        value,
        str,
    ):

        return bool(
            value.strip()
        )

    if isinstance(
        value,
        (
            list,
            dict,
            tuple,
            set,
        ),
    ):

        return bool(
            value
        )

    return True


# =========================================================
# Reload frozen Registry artifacts independently
# =========================================================

profile_claims = load_jsonl_records(
    registry_paths[
        "canonical_claims"
    ]
)

profile_all_evidence = load_jsonl_records(
    registry_paths[
        "evidence_links"
    ]
)

profile_relations = load_jsonl_records(
    registry_paths[
        "claim_relations"
    ]
)


assert (
    len(
        profile_claims
    )
    == registry_counts[
        "canonical_claims"
    ]
)


assert (
    len(
        profile_relations
    )
    == registry_counts[
        "claim_relations_total"
    ]
)


# =========================================================
# Claim identity validation
# =========================================================

profile_claim_by_id = {}


for claim in (
    profile_claims
):

    claim_id = claim.get(
        "claim_id"
    )

    assert (
        is_nonempty_value(
            claim_id
        )
    )

    assert (
        claim_id
        not in profile_claim_by_id
    ), (
        "Duplicate canonical claim_id: "
        f"{claim_id}"
    )

    profile_claim_by_id[
        claim_id
    ] = claim


assert (
    len(
        profile_claim_by_id
    )
    == 4006
)


# =========================================================
# Evidence identity validation
# =========================================================

profile_evidence_by_id = {}


for evidence in (
    profile_all_evidence
):

    evidence_id = evidence.get(
        "evidence_id"
    )

    assert (
        is_nonempty_value(
            evidence_id
        )
    )

    assert (
        evidence_id
        not in profile_evidence_by_id
    ), (
        "Duplicate evidence_id: "
        f"{evidence_id}"
    )

    profile_evidence_by_id[
        evidence_id
    ] = evidence


registry_evidence_ids = {
    evidence_id
    for claim in (
        profile_claims
    )
    for evidence_id in (
        claim[
            "evidence_ids"
        ]
    )
}


assert (
    len(
        registry_evidence_ids
    )
    == registry_counts[
        "validated_registry_evidence_occurrences"
    ]
)


missing_registry_evidence_ids = sorted(
    registry_evidence_ids
    - set(
        profile_evidence_by_id
    )
)


assert not missing_registry_evidence_ids, (
    "Canonical Registry references missing "
    "Evidence IDs."
)


profile_registry_evidence = [
    profile_evidence_by_id[
        evidence_id
    ]
    for evidence_id in sorted(
        registry_evidence_ids
    )
]


assert (
    len(
        profile_registry_evidence
    )
    == 4125
)


# =========================================================
# Structural eligibility
#
# IMPORTANT:
# This is NOT final Golden Seed eligibility.
# It checks only whether a Canonical Claim is structurally
# ready to ENTER Golden suitability review.
# =========================================================

structural_failures_by_claim = {}


for claim in (
    profile_claims
):

    claim_id = claim[
        "claim_id"
    ]

    failures = []


    if not is_nonempty_value(
        claim.get(
            "claim_text_fa"
        )
    ):

        failures.append(
            "MISSING_CLAIM_TEXT"
        )


    if (
        claim.get(
            "registry_validation_status"
        )
        != "VALIDATED"
    ):

        failures.append(
            "NOT_VALIDATED"
        )


    evidence_ids = claim.get(
        "evidence_ids",
        [],
    )


    if not evidence_ids:

        failures.append(
            "NO_EVIDENCE_IDS"
        )


    missing_evidence_for_claim = [
        evidence_id
        for evidence_id in (
            evidence_ids
        )
        if (
            evidence_id
            not in profile_evidence_by_id
        )
    ]


    if missing_evidence_for_claim:

        failures.append(
            "REFERENCES_MISSING_EVIDENCE"
        )


    if not claim.get(
        "source_ids"
    ):

        failures.append(
            "NO_SOURCE_IDS"
        )


    if not claim.get(
        "source_document_ids"
    ):

        failures.append(
            "NO_SOURCE_DOCUMENT_IDS"
        )


    if not claim.get(
        "source_types"
    ):

        failures.append(
            "NO_SOURCE_TYPES"
        )


    if not claim.get(
        "authority_classes"
    ):

        failures.append(
            "NO_AUTHORITY_CLASSES"
        )


    if not claim.get(
        "claim_discovery_roles"
    ):

        failures.append(
            "NO_DISCOVERY_ROLES"
        )


    if not claim.get(
        "claim_types"
    ):

        failures.append(
            "NO_CLAIM_TYPES"
        )


    if failures:

        structural_failures_by_claim[
            claim_id
        ] = failures


structurally_eligible_claim_ids = sorted(
    set(
        profile_claim_by_id
    )
    - set(
        structural_failures_by_claim
    )
)


structurally_ineligible_claim_ids = sorted(
    structural_failures_by_claim
)


# =========================================================
# Claim Type profile
# =========================================================

claim_type_membership_counts = Counter()

canonical_singular_claim_type_counts = Counter()


for claim in (
    profile_claims
):

    canonical_singular_claim_type_counts[
        str(
            claim.get(
                "claim_type"
            )
        )
    ] += 1


    for claim_type in (
        claim.get(
            "claim_types",
            []
        )
    ):

        claim_type_membership_counts[
            claim_type
        ] += 1


multi_claim_type_claims = [
    claim[
        "claim_id"
    ]
    for claim in (
        profile_claims
    )
    if (
        len(
            claim.get(
                "claim_types",
                []
            )
        )
        > 1
    )
]


# =========================================================
# Authority / discovery role / source type profiles
# =========================================================

authority_membership_counts = Counter()
discovery_role_membership_counts = Counter()
source_type_membership_counts = Counter()


multi_authority_claims = []
multi_discovery_role_claims = []
multi_source_type_claims = []


for claim in (
    profile_claims
):

    authorities = claim.get(
        "authority_classes",
        []
    )

    discovery_roles = claim.get(
        "claim_discovery_roles",
        []
    )

    source_types = claim.get(
        "source_types",
        []
    )


    for authority in authorities:

        authority_membership_counts[
            authority
        ] += 1


    for role in discovery_roles:

        discovery_role_membership_counts[
            role
        ] += 1


    for source_type in source_types:

        source_type_membership_counts[
            source_type
        ] += 1


    if len(
        authorities
    ) > 1:

        multi_authority_claims.append(
            claim[
                "claim_id"
            ]
        )


    if len(
        discovery_roles
    ) > 1:

        multi_discovery_role_claims.append(
            claim[
                "claim_id"
            ]
        )


    if len(
        source_types
    ) > 1:

        multi_source_type_claims.append(
            claim[
                "claim_id"
            ]
        )


# =========================================================
# Claim source / evidence multiplicity
# =========================================================

evidence_count_distribution = Counter(
    claim[
        "evidence_count"
    ]
    for claim in (
        profile_claims
    )
)


raw_claim_count_distribution = Counter(
    claim[
        "raw_claim_count"
    ]
    for claim in (
        profile_claims
    )
)


source_id_count_distribution = Counter(
    len(
        claim.get(
            "source_ids",
            []
        )
    )
    for claim in (
        profile_claims
    )
)


source_document_count_distribution = Counter(
    len(
        claim.get(
            "source_document_ids",
            []
        )
    )
    for claim in (
        profile_claims
    )
)


claims_with_multiple_evidence = sum(
    1
    for claim in (
        profile_claims
    )
    if (
        claim[
            "evidence_count"
        ]
        > 1
    )
)


claims_with_multiple_sources = sum(
    1
    for claim in (
        profile_claims
    )
    if (
        len(
            claim.get(
                "source_ids",
                []
            )
        )
        > 1
    )
)


claims_with_multiple_documents = sum(
    1
    for claim in (
        profile_claims
    )
    if (
        len(
            claim.get(
                "source_document_ids",
                []
            )
        )
        > 1
    )
)


# =========================================================
# Claim text length profile
# =========================================================

claim_text_lengths = [
    len(
        claim[
            "claim_text_fa"
        ]
    )
    for claim in (
        profile_claims
    )
]


claim_text_length_profile = (
    numeric_profile(
        claim_text_lengths
    )
)


claim_text_length_bands = Counter()


for length in (
    claim_text_lengths
):

    if length <= 40:

        band = (
            "LE_40"
        )

    elif length <= 80:

        band = (
            "41_TO_80"
        )

    elif length <= 160:

        band = (
            "81_TO_160"
        )

    elif length <= 320:

        band = (
            "161_TO_320"
        )

    elif length <= 640:

        band = (
            "321_TO_640"
        )

    else:

        band = (
            "GT_640"
        )


    claim_text_length_bands[
        band
    ] += 1


# =========================================================
# Evidence quality profile
# =========================================================

evidence_quote_lengths = []


evidence_missing_quote = 0
evidence_missing_locator = 0
evidence_missing_source_id = 0
evidence_missing_document_id = 0
evidence_missing_attribution = 0


for evidence in (
    profile_registry_evidence
):

    quote = evidence.get(
        "evidence_quote_exact"
    )


    if not is_nonempty_value(
        quote
    ):

        evidence_missing_quote += 1

    else:

        evidence_quote_lengths.append(
            len(
                quote
            )
        )


    if not is_nonempty_value(
        evidence.get(
            "locator"
        )
    ):

        evidence_missing_locator += 1


    if not is_nonempty_value(
        evidence.get(
            "source_id"
        )
    ):

        evidence_missing_source_id += 1


    if not is_nonempty_value(
        evidence.get(
            "document_id"
        )
    ):

        evidence_missing_document_id += 1


    if not is_nonempty_value(
        evidence.get(
            "attribution_snapshot"
        )
    ):

        evidence_missing_attribution += 1


assert (
    evidence_missing_quote
    == 0
), (
    "Validated Registry contains Evidence "
    "without exact quote."
)


assert (
    evidence_missing_source_id
    == 0
)


assert (
    evidence_missing_document_id
    == 0
)


evidence_quote_length_profile = (
    numeric_profile(
        evidence_quote_lengths
    )
)


# =========================================================
# Relation topology
# =========================================================

claim_relation_degree = Counter()

claim_relation_type_degree = {
    "SEMANTIC_DUPLICATE": Counter(),
    "SUBSUMES": Counter(),
    "PARTIAL_OVERLAP": Counter(),
}


graph_relations = [
    relation
    for relation in (
        profile_relations
    )
    if relation.get(
        "graph_edge_eligible"
    )
]


assert (
    len(
        graph_relations
    )
    == registry_counts[
        "graph_edge_eligible_claim_relations"
    ]
)


for relation in (
    graph_relations
):

    relation_type = relation[
        "relation_type"
    ]

    claim_id_a = relation[
        "claim_id_a"
    ]

    claim_id_b = relation[
        "claim_id_b"
    ]


    assert (
        claim_id_a
        in profile_claim_by_id
    )

    assert (
        claim_id_b
        in profile_claim_by_id
    )

    assert (
        claim_id_a
        != claim_id_b
    )


    claim_relation_degree[
        claim_id_a
    ] += 1

    claim_relation_degree[
        claim_id_b
    ] += 1


    assert (
        relation_type
        in claim_relation_type_degree
    )


    claim_relation_type_degree[
        relation_type
    ][
        claim_id_a
    ] += 1

    claim_relation_type_degree[
        relation_type
    ][
        claim_id_b
    ] += 1


claims_with_any_graph_relation = set(
    claim_relation_degree
)


claims_without_graph_relation = sorted(
    set(
        profile_claim_by_id
    )
    - claims_with_any_graph_relation
)


relation_degree_values = [
    claim_relation_degree.get(
        claim_id,
        0,
    )
    for claim_id in (
        profile_claim_by_id
    )
]


relation_degree_profile = numeric_profile(
    relation_degree_values
)


claims_with_semantic_duplicate = set(
    claim_relation_type_degree[
        "SEMANTIC_DUPLICATE"
    ]
)


claims_with_subsumes = set(
    claim_relation_type_degree[
        "SUBSUMES"
    ]
)


claims_with_partial_overlap = set(
    claim_relation_type_degree[
        "PARTIAL_OVERLAP"
    ]
)


# =========================================================
# Exact duplicate provenance profile
# =========================================================

exact_duplicate_collapsed_claims = [
    claim[
        "claim_id"
    ]
    for claim in (
        profile_claims
    )
    if claim.get(
        "exact_duplicate_collapsed"
    )
]


# =========================================================
# Structural failure counts
# =========================================================

structural_failure_reason_counts = Counter(
    reason
    for reasons in (
        structural_failures_by_claim.values()
    )
    for reason in reasons
)


# =========================================================
# Stage result
# =========================================================

claim_eligibility_profile_status = (
    "PASS"
)


golden_v2_eligibility_contract_readiness = (
    "READY_FOR_GOLDEN_V2_ELIGIBILITY_CONTRACT"
)


# =========================================================
# Result
# =========================================================

print(
    "=== NB12j REGISTRY CLAIM ELIGIBILITY PROFILE ==="
)

print()

print(
    "Canonical Claims:",
    len(
        profile_claims
    ),
)

print(
    "Structurally eligible for Golden review:",
    len(
        structurally_eligible_claim_ids
    ),
)

print(
    "Structurally ineligible:",
    len(
        structurally_ineligible_claim_ids
    ),
)

print(
    "Structural failure reasons:",
    sorted_counter_dict(
        structural_failure_reason_counts
    ),
)


print()

print(
    "=== CLAIM TYPES ==="
)

print(
    "Claim type membership counts:",
    sorted_counter_dict(
        claim_type_membership_counts
    ),
)

print(
    "Canonical singular claim_type counts:",
    sorted_counter_dict(
        canonical_singular_claim_type_counts
    ),
)

print(
    "Claims with multiple Claim Types:",
    len(
        multi_claim_type_claims
    ),
)


print()

print(
    "=== AUTHORITY / ROLE / SOURCE ==="
)

print(
    "Authority membership counts:",
    sorted_counter_dict(
        authority_membership_counts
    ),
)

print(
    "Discovery Role membership counts:",
    sorted_counter_dict(
        discovery_role_membership_counts
    ),
)

print(
    "Source Type membership counts:",
    sorted_counter_dict(
        source_type_membership_counts
    ),
)

print(
    "Multi-authority Claims:",
    len(
        multi_authority_claims
    ),
)

print(
    "Multi-role Claims:",
    len(
        multi_discovery_role_claims
    ),
)

print(
    "Multi-source-type Claims:",
    len(
        multi_source_type_claims
    ),
)


print()

print(
    "=== EVIDENCE / SOURCE MULTIPLICITY ==="
)

print(
    "Evidence count distribution:",
    sorted_counter_dict(
        evidence_count_distribution
    ),
)

print(
    "Raw Claim count distribution:",
    sorted_counter_dict(
        raw_claim_count_distribution
    ),
)

print(
    "Source ID count distribution:",
    sorted_counter_dict(
        source_id_count_distribution
    ),
)

print(
    "Source Document count distribution:",
    sorted_counter_dict(
        source_document_count_distribution
    ),
)

print(
    "Claims with >1 Evidence:",
    claims_with_multiple_evidence,
)

print(
    "Claims with >1 Source:",
    claims_with_multiple_sources,
)

print(
    "Claims with >1 Document:",
    claims_with_multiple_documents,
)


print()

print(
    "=== CLAIM TEXT LENGTH ==="
)

print(
    "Length profile:",
    claim_text_length_profile,
)

print(
    "Length bands:",
    sorted_counter_dict(
        claim_text_length_bands
    ),
)


print()

print(
    "=== VALIDATED EVIDENCE PROFILE ==="
)

print(
    "Registry Evidence occurrences:",
    len(
        profile_registry_evidence
    ),
)

print(
    "Evidence quote length profile:",
    evidence_quote_length_profile,
)

print(
    "Missing exact quote:",
    evidence_missing_quote,
)

print(
    "Missing locator:",
    evidence_missing_locator,
)

print(
    "Missing source_id:",
    evidence_missing_source_id,
)

print(
    "Missing document_id:",
    evidence_missing_document_id,
)

print(
    "Missing attribution snapshot:",
    evidence_missing_attribution,
)


print()

print(
    "=== RELATION TOPOLOGY ==="
)

print(
    "Graph-edge eligible Relations:",
    len(
        graph_relations
    ),
)

print(
    "Claims with any Graph Relation:",
    len(
        claims_with_any_graph_relation
    ),
)

print(
    "Claims without Graph Relation:",
    len(
        claims_without_graph_relation
    ),
)

print(
    "Relation degree profile:",
    relation_degree_profile,
)

print(
    "Claims with SEMANTIC_DUPLICATE relation:",
    len(
        claims_with_semantic_duplicate
    ),
)

print(
    "Claims with SUBSUMES relation:",
    len(
        claims_with_subsumes
    ),
)

print(
    "Claims with PARTIAL_OVERLAP relation:",
    len(
        claims_with_partial_overlap
    ),
)

print(
    "Canonical Claims with Exact-Duplicate provenance:",
    len(
        exact_duplicate_collapsed_claims
    ),
)


print()

print(
    "Final Golden Seed eligibility decided:",
    False,
)

print(
    "Fixed Golden Seed target invented:",
    False,
)

print(
    "Files modified by this Cell:",
    0,
)

print(
    "API calls performed:",
    0,
)


print()

print(
    "Next stage:",
    "GOLDEN_V2_ELIGIBILITY_CONTRACT",
)

print(
    "Readiness:",
    golden_v2_eligibility_contract_readiness,
)


print()

print(
    "CODE CELL 06 GATE:",
    claim_eligibility_profile_status,
)
````

<h3>اصل اساسی</h3>

<p>
هیچ Claim معتبر صرفاً به دلیل نوع Claim، Source Type، طول متن،
Primary/Secondary بودن یا داشتن Relation حذف نمی‌شود.
این ویژگی‌ها برای Routing، Review و Coverage استفاده می‌شوند.

<h3>Authority Policy</h3>

<ul>
<li><code>CORE_PRIMARY_KNOWLEDGE</code> وارد مسیر Standard Golden Review می‌شود.</li>
<li><code>SUPERVISED_SECONDARY_KNOWLEDGE</code> می‌تواند وارد Golden شود، اما Attribution آن باید حفظ شود.</li>
<li>Secondary Knowledge حق Override کردن Primary Evidence را ندارد.</li>
</ul>

<h3>Relation Policy</h3>

<ul>
<li><code>SEMANTIC_DUPLICATE</code> باعث حذف خودکار Claim نمی‌شود، اما Claim وارد Redundancy Review می‌شود.</li>
<li><code>SUBSUMES</code> باعث Merge خودکار نمی‌شود؛ Claimهای Parent و Child باید از نظر استقلال سؤال بررسی شوند.</li>
<li><code>PARTIAL_OVERLAP</code> باعث حذف نمی‌شود، اما برای جلوگیری از Question Redundancy علامت‌گذاری می‌شود.</li>
<li>Exact Duplicateها قبلاً در Registry Canonicalize شده‌اند و دیگر Seed مستقل جداگانه ایجاد نمی‌کنند.</li>
</ul>

<h3>Claim Type Policy</h3>

<p>
تمام Claim Typeهای مشاهده‌شده در Registry مجاز به ورود به Golden Suitability Review هستند.
در این مرحله هیچ Claim Type با Threshold دلخواه حذف نمی‌شود.
</p>

<h3>Text Length Policy</h3>

<p>
طول Claim Text یک Diagnostic Feature است و Hard Exclusion ایجاد نمی‌کند.
Claim کوتاه یا بلند فقط در مرحله Question Generation و Quality Review
باید از نظر قابلیت ساخت سؤال مستقل بررسی شود.
</p>

<h3>Final Seed Requirement</h3>

<p>
یک Claim فقط زمانی Final Golden Seed محسوب خواهد شد که علاوه بر Structural Eligibility،
بتوان از آن یک سؤال مستقل، غیرمبهم، Evidence-Grounded و قابل پاسخ از Evidence واقعی ساخت.
این تصمیم در مراحل بعد انجام می‌شود و در این مرحله هیچ تعداد Seed هدف اختراع نمی‌شود.
</p>

````python
# Code Cell 09

assert (
    claim_eligibility_manifest_status
    == "PASS"
), (
    "Code Cell 08 must PASS before "
    "Seed Suitability Discovery."
)

assert (
    seed_suitability_discovery_readiness
    == "READY_FOR_SEED_SUITABILITY_DISCOVERY"
)


# =========================================================
# Graph helper
# =========================================================

def connected_components_from_edges(
    node_ids,
    edges,
):

    adjacency = defaultdict(
        set
    )


    for (
        node_a,
        node_b,
    ) in edges:

        assert (
            node_a
            in node_ids
        )

        assert (
            node_b
            in node_ids
        )

        assert (
            node_a
            != node_b
        )


        adjacency[
            node_a
        ].add(
            node_b
        )

        adjacency[
            node_b
        ].add(
            node_a
        )


    components = []

    visited = set()


    for start_node in sorted(
        adjacency
    ):

        if (
            start_node
            in visited
        ):
            continue


        stack = [
            start_node
        ]

        component = []


        while stack:

            current = stack.pop()


            if (
                current
                in visited
            ):
                continue


            visited.add(
                current
            )

            component.append(
                current
            )


            for neighbor in sorted(
                adjacency[
                    current
                ],
                reverse=True,
            ):

                if (
                    neighbor
                    not in visited
                ):

                    stack.append(
                        neighbor
                    )


        components.append(
            sorted(
                component
            )
        )


    components = sorted(
        components,
        key=lambda component: (
            -len(
                component
            ),
            component[
                0
            ],
        ),
    )


    return components


# =========================================================
# Split adjudicated Graph Relations by type
# =========================================================

semantic_duplicate_edges = []

subsumes_edges = []

partial_overlap_edges = []


for relation in (
    graph_relations
):

    relation_type = relation[
        "relation_type"
    ]

    claim_id_a = relation[
        "claim_id_a"
    ]

    claim_id_b = relation[
        "claim_id_b"
    ]


    if (
        relation_type
        == "SEMANTIC_DUPLICATE"
    ):

        semantic_duplicate_edges.append(
            (
                claim_id_a,
                claim_id_b,
            )
        )


    elif (
        relation_type
        == "SUBSUMES"
    ):

        subsumes_edges.append(
            (
                claim_id_a,
                claim_id_b,
            )
        )


    elif (
        relation_type
        == "PARTIAL_OVERLAP"
    ):

        partial_overlap_edges.append(
            (
                claim_id_a,
                claim_id_b,
            )
        )


    else:

        raise AssertionError(
            "Unexpected graph relation type: "
            f"{relation_type}"
        )


assert (
    len(
        semantic_duplicate_edges
    )
    == 233
)


assert (
    len(
        subsumes_edges
    )
    == 544
)


assert (
    len(
        partial_overlap_edges
    )
    == 1638
)


# =========================================================
# Semantic Duplicate connected components
# =========================================================

all_claim_ids = set(
    profile_claim_by_id
)


semantic_duplicate_components = (
    connected_components_from_edges(
        all_claim_ids,
        semantic_duplicate_edges,
    )
)


semantic_duplicate_component_members = {
    claim_id
    for component in (
        semantic_duplicate_components
    )
    for claim_id in (
        component
    )
}


assert (
    semantic_duplicate_component_members
    == claims_with_semantic_duplicate
)


semantic_duplicate_component_size_counts = Counter(
    len(
        component
    )
    for component in (
        semantic_duplicate_components
    )
)


semantic_duplicate_component_count = len(
    semantic_duplicate_components
)


semantic_duplicate_member_count = len(
    semantic_duplicate_component_members
)


semantic_duplicate_redundant_claim_slots = sum(
    len(
        component
    )
    - 1
    for component in (
        semantic_duplicate_components
    )
)


assert (
    semantic_duplicate_redundant_claim_slots
    == (
        semantic_duplicate_member_count
        - semantic_duplicate_component_count
    )
)


# =========================================================
# Theoretical duplicate-controlled candidate universe
#
# Important:
# This does NOT modify Registry identity.
#
# For Golden only:
# every Semantic-Duplicate connected component contributes
# at most one independent semantic slot before question
# generation.
# =========================================================

claims_outside_semantic_duplicate = (
    len(
        profile_claims
    )
    - semantic_duplicate_member_count
)


duplicate_controlled_candidate_ceiling = (
    claims_outside_semantic_duplicate
    + semantic_duplicate_component_count
)


assert (
    duplicate_controlled_candidate_ceiling
    == (
        len(
            profile_claims
        )
        - semantic_duplicate_redundant_claim_slots
    )
)


# =========================================================
# Inspect Duplicate Component authority composition
# =========================================================

semantic_duplicate_component_authority_patterns = Counter()

semantic_duplicate_mixed_authority_components = []

semantic_duplicate_single_authority_components = []


for (
    component_index,
    component,
) in enumerate(
    semantic_duplicate_components,
    start=1,
):

    authorities = sorted({
        profile_claim_by_id[
            claim_id
        ][
            "authority_classes"
        ][
            0
        ]
        for claim_id in component
    })


    authority_pattern = (
        " + ".join(
            authorities
        )
    )


    semantic_duplicate_component_authority_patterns[
        authority_pattern
    ] += 1


    component_record = {
        "component_index": (
            component_index
        ),

        "size": (
            len(
                component
            )
        ),

        "claim_ids": (
            component
        ),

        "authorities": (
            authorities
        ),
    }


    if (
        len(
            authorities
        )
        > 1
    ):

        semantic_duplicate_mixed_authority_components.append(
            component_record
        )

    else:

        semantic_duplicate_single_authority_components.append(
            component_record
        )


# =========================================================
# Duplicate Component source-type composition
# =========================================================

semantic_duplicate_component_source_patterns = Counter()

semantic_duplicate_mixed_source_type_components = []


for (
    component_index,
    component,
) in enumerate(
    semantic_duplicate_components,
    start=1,
):

    source_types_in_component = sorted({
        source_type
        for claim_id in component
        for source_type in (
            profile_claim_by_id[
                claim_id
            ][
                "source_types"
            ]
        )
    })


    pattern = (
        " + ".join(
            source_types_in_component
        )
    )


    semantic_duplicate_component_source_patterns[
        pattern
    ] += 1


    if (
        len(
            source_types_in_component
        )
        > 1
    ):

        semantic_duplicate_mixed_source_type_components.append({
            "component_index": (
                component_index
            ),

            "size": (
                len(
                    component
                )
            ),

            "source_types": (
                source_types_in_component
            ),

            "claim_ids": (
                component
            ),
        })


# =========================================================
# Relation membership combinations
# =========================================================

relation_membership_combination_counts = Counter()

relation_membership_claim_ids = defaultdict(
    list
)


for claim_id in sorted(
    all_claim_ids
):

    memberships = []


    if (
        claim_id
        in claims_with_semantic_duplicate
    ):

        memberships.append(
            "SEMANTIC_DUPLICATE"
        )


    if (
        claim_id
        in claims_with_subsumes
    ):

        memberships.append(
            "SUBSUMES"
        )


    if (
        claim_id
        in claims_with_partial_overlap
    ):

        memberships.append(
            "PARTIAL_OVERLAP"
        )


    if not memberships:

        combination = (
            "NONE"
        )

    else:

        combination = (
            " + ".join(
                memberships
            )
        )


    relation_membership_combination_counts[
        combination
    ] += 1


    relation_membership_claim_ids[
        combination
    ].append(
        claim_id
    )


assert (
    sum(
        relation_membership_combination_counts.values()
    )
    == 4006
)


assert (
    relation_membership_combination_counts[
        "NONE"
    ]
    == 2743
)


# =========================================================
# Review populations after Semantic-Duplicate control
# =========================================================

claims_needing_duplicate_representative_selection = (
    semantic_duplicate_component_members
)


claims_with_subsumes_not_semantic_duplicate = (
    claims_with_subsumes
    - claims_with_semantic_duplicate
)


claims_with_partial_overlap_not_semantic_duplicate = (
    claims_with_partial_overlap
    - claims_with_semantic_duplicate
)


claims_with_overlap_or_subsumption_after_duplicate_control = (
    (
        claims_with_subsumes
        | claims_with_partial_overlap
    )
    - claims_with_semantic_duplicate
)


# =========================================================
# All semantic-relation connected components
#
# This is diagnostic only.
# It shows how large the overall relation neighborhoods are.
# =========================================================

all_semantic_relation_edges = (
    semantic_duplicate_edges
    + subsumes_edges
    + partial_overlap_edges
)


all_relation_components = (
    connected_components_from_edges(
        all_claim_ids,
        all_semantic_relation_edges,
    )
)


all_relation_component_size_counts = Counter(
    len(
        component
    )
    for component in (
        all_relation_components
    )
)


all_relation_component_members = {
    claim_id
    for component in (
        all_relation_components
    )
    for claim_id in (
        component
    )
}


assert (
    all_relation_component_members
    == claims_with_any_graph_relation
)


largest_relation_component_size = (
    max(
        (
            len(
                component
            )
            for component in (
                all_relation_components
            )
        ),
        default=0,
    )
)


largest_semantic_duplicate_component_size = (
    max(
        (
            len(
                component
            )
            for component in (
                semantic_duplicate_components
            )
        ),
        default=0,
    )
)


# =========================================================
# Claim-Type profile after theoretical Duplicate control
#
# No representative is chosen yet, so we report:
# - claims outside duplicate components
# - duplicate components by member Claim Types
# =========================================================

claims_outside_duplicate_by_type = Counter()


for claim in (
    profile_claims
):

    if (
        claim[
            "claim_id"
        ]
        in semantic_duplicate_component_members
    ):

        continue


    for claim_type in (
        claim[
            "claim_types"
        ]
    ):

        claims_outside_duplicate_by_type[
            claim_type
        ] += 1


duplicate_component_claim_type_patterns = Counter()


for component in (
    semantic_duplicate_components
):

    types = sorted({
        claim_type
        for claim_id in component
        for claim_type in (
            profile_claim_by_id[
                claim_id
            ][
                "claim_types"
            ]
        )
    })


    duplicate_component_claim_type_patterns[
        " + ".join(
            types
        )
    ] += 1


# =========================================================
# Cost / adjudication boundary
#
# Relation adjudication is already complete in NB12i.
# What remains is Seed-selection policy and later
# Question-level independence / quality validation.
# =========================================================

semantic_relation_adjudication_already_complete = (
    True
)


new_semantic_relation_adjudication_required_now = (
    False
)


duplicate_component_representative_decisions_required = (
    semantic_duplicate_component_count
)


subsumes_pairs_for_question_independence_awareness = (
    len(
        subsumes_edges
    )
)


partial_overlap_pairs_for_question_redundancy_awareness = (
    len(
        partial_overlap_edges
    )
)


# =========================================================
# Stage Gate
# =========================================================

seed_suitability_discovery_status = (
    "PASS"
)


seed_selection_contract_readiness = (
    "READY_FOR_SEED_SELECTION_CONTRACT"
)


# =========================================================
# Result
# =========================================================

print(
    "=== NB12j SEED SUITABILITY DISCOVERY ==="
)

print()

print(
    "Canonical structurally eligible Claims:",
    len(
        profile_claims
    ),
)


print()

print(
    "=== SEMANTIC DUPLICATE GROUPS ==="
)

print(
    "SEMANTIC_DUPLICATE relations:",
    len(
        semantic_duplicate_edges
    ),
)

print(
    "Claims inside Semantic-Duplicate components:",
    semantic_duplicate_member_count,
)

print(
    "Semantic-Duplicate connected components:",
    semantic_duplicate_component_count,
)

print(
    "Component size distribution:",
    dict(
        sorted(
            semantic_duplicate_component_size_counts.items()
        )
    ),
)

print(
    "Largest Semantic-Duplicate component:",
    largest_semantic_duplicate_component_size,
)

print(
    "Redundant Claim slots if one semantic slot "
    "is kept per component:",
    semantic_duplicate_redundant_claim_slots,
)


print()

print(
    "Claims outside Semantic-Duplicate components:",
    claims_outside_semantic_duplicate,
)

print(
    "Duplicate-controlled candidate ceiling:",
    duplicate_controlled_candidate_ceiling,
)


print()

print(
    "Semantic-Duplicate authority patterns:",
    dict(
        sorted(
            semantic_duplicate_component_authority_patterns.items()
        )
    ),
)

print(
    "Mixed-authority Semantic-Duplicate components:",
    len(
        semantic_duplicate_mixed_authority_components
    ),
)

print(
    "Semantic-Duplicate source-type patterns:",
    dict(
        sorted(
            semantic_duplicate_component_source_patterns.items()
        )
    ),
)

print(
    "Mixed-source-type Semantic-Duplicate components:",
    len(
        semantic_duplicate_mixed_source_type_components
    ),
)


print()

print(
    "=== RELATION MEMBERSHIP COMBINATIONS ==="
)

for (
    combination,
    count,
) in sorted(
    relation_membership_combination_counts.items()
):

    print(
        "  ",
        combination,
        ":",
        count,
    )


print()

print(
    "Claims needing Duplicate representative selection:",
    len(
        claims_needing_duplicate_representative_selection
    ),
)

print(
    "Duplicate components requiring one representative decision:",
    duplicate_component_representative_decisions_required,
)

print(
    "Claims with SUBSUMES but no Semantic-Duplicate flag:",
    len(
        claims_with_subsumes_not_semantic_duplicate
    ),
)

print(
    "Claims with PARTIAL_OVERLAP but no Semantic-Duplicate flag:",
    len(
        claims_with_partial_overlap_not_semantic_duplicate
    ),
)

print(
    "Claims with SUBSUMES and/or PARTIAL_OVERLAP "
    "after excluding Semantic-Duplicate members:",
    len(
        claims_with_overlap_or_subsumption_after_duplicate_control
    ),
)


print()

print(
    "=== OVERALL RELATION NEIGHBORHOODS ==="
)

print(
    "Claims in any semantic relation neighborhood:",
    len(
        all_relation_component_members
    ),
)

print(
    "All-relation connected components:",
    len(
        all_relation_components
    ),
)

print(
    "All-relation component size distribution:",
    dict(
        sorted(
            all_relation_component_size_counts.items()
        )
    ),
)

print(
    "Largest all-relation component:",
    largest_relation_component_size,
)


print()

print(
    "=== ADJUDICATION BOUNDARY ==="
)

print(
    "Semantic Relation adjudication already complete:",
    semantic_relation_adjudication_already_complete,
)

print(
    "New Semantic Relation adjudication required now:",
    new_semantic_relation_adjudication_required_now,
)

print(
    "SUBSUMES pairs to respect during Question design:",
    subsumes_pairs_for_question_independence_awareness,
)

print(
    "PARTIAL_OVERLAP pairs to respect during Question design:",
    partial_overlap_pairs_for_question_redundancy_awareness,
)


print()

print(
    "Fixed final Seed target decided:",
    False,
)

print(
    "Seed representatives selected:",
    False,
)

print(
    "Question generation started:",
    False,
)

print(
    "Files modified by this Cell:",
    0,
)

print(
    "API calls performed:",
    0,
)


print()

print(
    "Next stage:",
    "SEED_SELECTION_CONTRACT",
)

print(
    "Readiness:",
    seed_selection_contract_readiness,
)


print()

print(
    "CODE CELL 09 GATE:",
    seed_suitability_discovery_status,
)
````

````python
# Code Cell 11

assert (
    seed_selection_contract_status
    == "PASS"
), (
    "Code Cell 10 must PASS before "
    "Provisional Seed Slot Construction."
)

assert (
    provisional_seed_slot_readiness
    == "READY_FOR_PROVISIONAL_SEED_SLOT_CONSTRUCTION"
)


# =========================================================
# Configuration
# =========================================================

PROVISIONAL_SEED_POOL_VERSION = (
    "golden_v2_seed_pool_v1"
)


provisional_seed_pool_path = (
    golden_v2_artifact_root
    / golden_v2_artifact_family[
        "seed_pool"
    ]
)


# =========================================================
# Representative ranking
# =========================================================

def representative_rank(
    claim_id,
):

    claim = (
        profile_claim_by_id[
            claim_id
        ]
    )


    authority_class = (
        claim[
            "authority_classes"
        ][
            0
        ]
    )


    return (
        representative_authority_priority[
            authority_class
        ],

        -claim[
            "evidence_count"
        ],

        -len(
            claim[
                "source_document_ids"
            ]
        ),

        -len(
            claim[
                "source_ids"
            ]
        ),

        claim_id,
    )


# =========================================================
# Build semantic units
# =========================================================

semantic_duplicate_member_set = set(
    semantic_duplicate_component_members
)


singleton_member_sets = [
    [
        claim_id
    ]
    for claim_id in sorted(
        all_claim_ids
        - semantic_duplicate_member_set
    )
]


duplicate_member_sets = [
    sorted(
        component
    )
    for component in (
        semantic_duplicate_components
    )
]


assert (
    len(
        singleton_member_sets
    )
    == 3705
)


assert (
    len(
        duplicate_member_sets
    )
    == 112
)


all_seed_member_sets = (
    singleton_member_sets
    + duplicate_member_sets
)


assert (
    len(
        all_seed_member_sets
    )
    == 3817
)


# =========================================================
# Stable Seed Slot IDs
# =========================================================

def make_seed_slot_id(
    member_claim_ids,
):

    payload = {
        "policy_version": (
            GOLDEN_V2_SEED_SELECTION_POLICY_VERSION
        ),

        "member_claim_ids": sorted(
            member_claim_ids
        ),
    }


    payload_text = json.dumps(
        payload,
        ensure_ascii=False,
        sort_keys=True,
        separators=(
            ",",
            ":",
        ),
    )


    digest = hashlib.sha256(
        payload_text.encode(
            "utf-8"
        )
    ).hexdigest()


    return (
        "gsv2_slot_"
        + digest[
            :24
        ]
    )


# =========================================================
# Build preliminary Slot records
# =========================================================

preliminary_seed_slots = []

claim_to_seed_slot_id = {}


for member_claim_ids in (
    all_seed_member_sets
):

    member_claim_ids = sorted(
        member_claim_ids
    )


    representative_claim_id = min(
        member_claim_ids,
        key=representative_rank,
    )


    representative_claim = (
        profile_claim_by_id[
            representative_claim_id
        ]
    )


    seed_slot_id = (
        make_seed_slot_id(
            member_claim_ids
        )
    )


    assert (
        seed_slot_id
        not in {
            record[
                "seed_slot_id"
            ]
            for record in (
                preliminary_seed_slots
            )
        }
    ), (
        "Seed Slot ID collision: "
        f"{seed_slot_id}"
    )


    for claim_id in (
        member_claim_ids
    ):

        assert (
            claim_id
            not in claim_to_seed_slot_id
        )

        claim_to_seed_slot_id[
            claim_id
        ] = seed_slot_id


    member_claims = [
        profile_claim_by_id[
            claim_id
        ]
        for claim_id in (
            member_claim_ids
        )
    ]


    member_authorities = sorted({
        claim[
            "authority_classes"
        ][
            0
        ]
        for claim in (
            member_claims
        )
    })


    member_source_types = sorted({
        source_type
        for claim in (
            member_claims
        )
        for source_type in (
            claim[
                "source_types"
            ]
        )
    })


    member_claim_types = sorted({
        claim_type
        for claim in (
            member_claims
        )
        for claim_type in (
            claim[
                "claim_types"
            ]
        )
    })


    component_evidence_ids = sorted({
        evidence_id
        for claim in (
            member_claims
        )
        for evidence_id in (
            claim[
                "evidence_ids"
            ]
        )
    })


    component_source_ids = sorted({
        source_id
        for claim in (
            member_claims
        )
        for source_id in (
            claim[
                "source_ids"
            ]
        )
    })


    component_document_ids = sorted({
        document_id
        for claim in (
            member_claims
        )
        for document_id in (
            claim[
                "source_document_ids"
            ]
        )
    })


    nonrepresentative_claim_ids = [
        claim_id
        for claim_id in (
            member_claim_ids
        )
        if (
            claim_id
            != representative_claim_id
        )
    ]


    is_semantic_duplicate_component = (
        len(
            member_claim_ids
        )
        > 1
    )


    preliminary_seed_slots.append({
        "seed_pool_version": (
            PROVISIONAL_SEED_POOL_VERSION
        ),

        "seed_selection_policy_version": (
            GOLDEN_V2_SEED_SELECTION_POLICY_VERSION
        ),

        "seed_selection_policy_sha256": (
            golden_v2_seed_selection_policy_sha256
        ),

        "seed_slot_id": (
            seed_slot_id
        ),

        "slot_kind": (
            "SEMANTIC_DUPLICATE_COMPONENT"
            if is_semantic_duplicate_component
            else "SINGLE_CANONICAL_CLAIM"
        ),

        "member_claim_ids": (
            member_claim_ids
        ),

        "member_claim_count": (
            len(
                member_claim_ids
            )
        ),

        "representative_claim_id": (
            representative_claim_id
        ),

        "nonrepresentative_claim_ids": (
            nonrepresentative_claim_ids
        ),

        "representative_selection_mode": (
            "DETERMINISTIC_RANKING"
            if is_semantic_duplicate_component
            else "SINGLETON_IDENTITY"
        ),

        "representative_claim": {
            "claim_text_fa": (
                representative_claim[
                    "claim_text_fa"
                ]
            ),

            "claim_text_sha256": (
                representative_claim[
                    "claim_text_sha256"
                ]
            ),

            "claim_types": (
                representative_claim[
                    "claim_types"
                ]
            ),

            "authority_class": (
                representative_claim[
                    "authority_classes"
                ][
                    0
                ]
            ),

            "source_types": (
                representative_claim[
                    "source_types"
                ]
            ),

            "source_ids": (
                representative_claim[
                    "source_ids"
                ]
            ),

            "source_document_ids": (
                representative_claim[
                    "source_document_ids"
                ]
            ),

            "evidence_ids": (
                representative_claim[
                    "evidence_ids"
                ]
            ),

            "evidence_count": (
                representative_claim[
                    "evidence_count"
                ]
            ),
        },

        "component_profile": {
            "authority_classes": (
                member_authorities
            ),

            "source_types": (
                member_source_types
            ),

            "claim_types": (
                member_claim_types
            ),

            "component_evidence_ids": (
                component_evidence_ids
            ),

            "component_evidence_count": (
                len(
                    component_evidence_ids
                )
            ),

            "component_source_ids": (
                component_source_ids
            ),

            "component_source_count": (
                len(
                    component_source_ids
                )
            ),

            "component_document_ids": (
                component_document_ids
            ),

            "component_document_count": (
                len(
                    component_document_ids
                )
            ),

            "mixed_authority": (
                len(
                    member_authorities
                )
                > 1
            ),

            "mixed_source_type": (
                len(
                    member_source_types
                )
                > 1
            ),
        },

        "evidence_selection_status": (
            "NOT_STARTED"
        ),

        "semantic_suitability_status": (
            "PENDING_SEMANTIC_SUITABILITY_REVIEW"
        ),

        "final_seed_selected": (
            False
        ),

        "question_generation_started": (
            False
        ),
    })


# =========================================================
# Validate total Claim preservation
# =========================================================

assert (
    len(
        preliminary_seed_slots
    )
    == 3817
)


assert (
    len(
        claim_to_seed_slot_id
    )
    == 4006
)


assert (
    set(
        claim_to_seed_slot_id
    )
    == all_claim_ids
)


all_member_claim_occurrences = sum(
    slot[
        "member_claim_count"
    ]
    for slot in (
        preliminary_seed_slots
    )
)


assert (
    all_member_claim_occurrences
    == 4006
)


nonrepresentative_member_count = sum(
    len(
        slot[
            "nonrepresentative_claim_ids"
        ]
    )
    for slot in (
        preliminary_seed_slots
    )
)


assert (
    nonrepresentative_member_count
    == semantic_duplicate_redundant_claim_slots
)


assert (
    nonrepresentative_member_count
    == 189
)


# =========================================================
# Validate Representative policy
# =========================================================

duplicate_slots = [
    slot
    for slot in (
        preliminary_seed_slots
    )
    if (
        slot[
            "slot_kind"
        ]
        == "SEMANTIC_DUPLICATE_COMPONENT"
    )
]


singleton_slots = [
    slot
    for slot in (
        preliminary_seed_slots
    )
    if (
        slot[
            "slot_kind"
        ]
        == "SINGLE_CANONICAL_CLAIM"
    )
]


assert (
    len(
        duplicate_slots
    )
    == 112
)


assert (
    len(
        singleton_slots
    )
    == 3705
)


for slot in (
    duplicate_slots
):

    expected_representative = min(
        slot[
            "member_claim_ids"
        ],
        key=representative_rank,
    )


    assert (
        slot[
            "representative_claim_id"
        ]
        == expected_representative
    )


# Mixed-authority Duplicate Components must select Primary.
mixed_authority_duplicate_slots = [
    slot
    for slot in (
        duplicate_slots
    )
    if (
        slot[
            "component_profile"
        ][
            "mixed_authority"
        ]
    )
]


assert (
    len(
        mixed_authority_duplicate_slots
    )
    == 23
)


assert all(
    slot[
        "representative_claim"
    ][
        "authority_class"
    ]
    == "CORE_PRIMARY_KNOWLEDGE"
    for slot in (
        mixed_authority_duplicate_slots
    )
)


# =========================================================
# Relation index
# =========================================================

relation_by_id = {
    relation[
        "relation_id"
    ]: relation
    for relation in (
        graph_relations
    )
}


assert (
    len(
        relation_by_id
    )
    == len(
        graph_relations
    )
)


# =========================================================
# Project Claim Relations onto Seed Slots
# =========================================================

slot_relation_profiles = {
    slot[
        "seed_slot_id"
    ]: {
        "internal_relation_ids": {
            "SEMANTIC_DUPLICATE": [],
            "SUBSUMES": [],
            "PARTIAL_OVERLAP": [],
        },

        "external_relation_ids": {
            "SEMANTIC_DUPLICATE": [],
            "SUBSUMES": [],
            "PARTIAL_OVERLAP": [],
        },

        "external_neighbor_seed_slot_ids": {
            "SEMANTIC_DUPLICATE": set(),
            "SUBSUMES": set(),
            "PARTIAL_OVERLAP": set(),
        },
    }
    for slot in (
        preliminary_seed_slots
    )
}


internal_projected_relation_counts = Counter()

external_projected_relation_counts = Counter()


for relation in (
    graph_relations
):

    relation_id = (
        relation[
            "relation_id"
        ]
    )

    relation_type = (
        relation[
            "relation_type"
        ]
    )

    claim_id_a = (
        relation[
            "claim_id_a"
        ]
    )

    claim_id_b = (
        relation[
            "claim_id_b"
        ]
    )


    slot_id_a = (
        claim_to_seed_slot_id[
            claim_id_a
        ]
    )

    slot_id_b = (
        claim_to_seed_slot_id[
            claim_id_b
        ]
    )


    if (
        slot_id_a
        == slot_id_b
    ):

        internal_projected_relation_counts[
            relation_type
        ] += 1


        slot_relation_profiles[
            slot_id_a
        ][
            "internal_relation_ids"
        ][
            relation_type
        ].append(
            relation_id
        )


    else:

        external_projected_relation_counts[
            relation_type
        ] += 1


        for (
            source_slot_id,
            neighbor_slot_id,
        ) in [
            (
                slot_id_a,
                slot_id_b,
            ),
            (
                slot_id_b,
                slot_id_a,
            ),
        ]:

            slot_relation_profiles[
                source_slot_id
            ][
                "external_relation_ids"
            ][
                relation_type
            ].append(
                relation_id
            )


            slot_relation_profiles[
                source_slot_id
            ][
                "external_neighbor_seed_slot_ids"
            ][
                relation_type
            ].add(
                neighbor_slot_id
            )


# Every Semantic-Duplicate relation must collapse inside
# its own Semantic Duplicate component.
assert (
    external_projected_relation_counts[
        "SEMANTIC_DUPLICATE"
    ]
    == 0
)


assert (
    internal_projected_relation_counts[
        "SEMANTIC_DUPLICATE"
    ]
    == len(
        semantic_duplicate_edges
    )
)


assert (
    internal_projected_relation_counts[
        "SEMANTIC_DUPLICATE"
    ]
    == 233
)


# =========================================================
# Finalize Seed Slot records with projected relations
# =========================================================

provisional_seed_slot_records = []


for slot in (
    preliminary_seed_slots
):

    seed_slot_id = (
        slot[
            "seed_slot_id"
        ]
    )


    relation_profile = (
        slot_relation_profiles[
            seed_slot_id
        ]
    )


    external_subsumes_count = len(
        relation_profile[
            "external_relation_ids"
        ][
            "SUBSUMES"
        ]
    )


    external_partial_overlap_count = len(
        relation_profile[
            "external_relation_ids"
        ][
            "PARTIAL_OVERLAP"
        ]
    )


    review_flags = []


    if (
        slot[
            "slot_kind"
        ]
        == "SEMANTIC_DUPLICATE_COMPONENT"
    ):

        review_flags.append(
            "SEMANTIC_DUPLICATE_COMPONENT"
        )


    if (
        slot[
            "component_profile"
        ][
            "mixed_authority"
        ]
    ):

        review_flags.append(
            "MIXED_AUTHORITY_COMPONENT"
        )


    if (
        slot[
            "component_profile"
        ][
            "mixed_source_type"
        ]
    ):

        review_flags.append(
            "MIXED_SOURCE_TYPE_COMPONENT"
        )


    if (
        slot[
            "representative_claim"
        ][
            "authority_class"
        ]
        == "SUPERVISED_SECONDARY_KNOWLEDGE"
    ):

        review_flags.append(
            "ATTRIBUTION_REQUIRED"
        )


    if (
        external_subsumes_count
        > 0
    ):

        review_flags.append(
            "SUBSUMPTION_REVIEW"
        )


    if (
        external_partial_overlap_count
        > 0
    ):

        review_flags.append(
            "PARTIAL_OVERLAP_REVIEW"
        )


    if (
        len(
            slot[
                "representative_claim"
            ][
                "claim_text_fa"
            ]
        )
        <= 40
    ):

        review_flags.append(
            "SHORT_TEXT_DIAGNOSTIC"
        )


    relation_projection = {
        "internal_relation_ids": {
            relation_type: sorted(
                relation_ids
            )
            for (
                relation_type,
                relation_ids,
            ) in (
                relation_profile[
                    "internal_relation_ids"
                ].items()
            )
        },

        "external_relation_ids": {
            relation_type: sorted(
                relation_ids
            )
            for (
                relation_type,
                relation_ids,
            ) in (
                relation_profile[
                    "external_relation_ids"
                ].items()
            )
        },

        "external_neighbor_seed_slot_ids": {
            relation_type: sorted(
                neighbor_ids
            )
            for (
                relation_type,
                neighbor_ids,
            ) in (
                relation_profile[
                    "external_neighbor_seed_slot_ids"
                ].items()
            )
        },

        "external_subsumes_relation_count": (
            external_subsumes_count
        ),

        "external_partial_overlap_relation_count": (
            external_partial_overlap_count
        ),

        "relation_aware_review_required": (
            (
                external_subsumes_count
                > 0
            )
            or (
                external_partial_overlap_count
                > 0
            )
        ),
    }


    final_record = dict(
        slot
    )


    final_record[
        "relation_projection"
    ] = relation_projection


    final_record[
        "review_flags"
    ] = sorted(
        review_flags
    )


    final_record[
        "review_flag_count"
    ] = len(
        review_flags
    )


    final_record[
        "seed_pool_status"
    ] = (
        "PROVISIONAL_SEED_SLOT"
    )


    provisional_seed_slot_records.append(
        final_record
    )


# =========================================================
# Deterministic ordering
# =========================================================

provisional_seed_slot_records = sorted(
    provisional_seed_slot_records,
    key=lambda record: (
        record[
            "representative_claim_id"
        ],
        record[
            "seed_slot_id"
        ],
    ),
)


# =========================================================
# Slot-level validation
# =========================================================

assert (
    len(
        provisional_seed_slot_records
    )
    == 3817
)


seed_slot_ids = [
    record[
        "seed_slot_id"
    ]
    for record in (
        provisional_seed_slot_records
    )
]


assert (
    len(
        seed_slot_ids
    )
    == len(
        set(
            seed_slot_ids
        )
    )
)


representative_claim_ids = [
    record[
        "representative_claim_id"
    ]
    for record in (
        provisional_seed_slot_records
    )
]


assert (
    len(
        representative_claim_ids
    )
    == len(
        set(
            representative_claim_ids
        )
    )
)


assert all(
    record[
        "final_seed_selected"
    ]
    is False
    for record in (
        provisional_seed_slot_records
    )
)


assert all(
    record[
        "semantic_suitability_status"
    ]
    == "PENDING_SEMANTIC_SUITABILITY_REVIEW"
    for record in (
        provisional_seed_slot_records
    )
)


assert all(
    record[
        "question_generation_started"
    ]
    is False
    for record in (
        provisional_seed_slot_records
    )
)


# =========================================================
# Slot-level diagnostics
# =========================================================

representative_authority_counts = Counter(
    record[
        "representative_claim"
    ][
        "authority_class"
    ]
    for record in (
        provisional_seed_slot_records
    )
)


representative_source_type_counts = Counter(
    source_type
    for record in (
        provisional_seed_slot_records
    )
    for source_type in (
        record[
            "representative_claim"
        ][
            "source_types"
        ]
    )
)


representative_claim_type_counts = Counter(
    claim_type
    for record in (
        provisional_seed_slot_records
    )
    for claim_type in (
        record[
            "representative_claim"
        ][
            "claim_types"
        ]
    )
)


slot_review_flag_counts = Counter(
    flag
    for record in (
        provisional_seed_slot_records
    )
    for flag in (
        record[
            "review_flags"
        ]
    )
)


relation_aware_seed_slots = [
    record[
        "seed_slot_id"
    ]
    for record in (
        provisional_seed_slot_records
    )
    if (
        record[
            "relation_projection"
        ][
            "relation_aware_review_required"
        ]
    )
]


relation_isolated_seed_slots = [
    record[
        "seed_slot_id"
    ]
    for record in (
        provisional_seed_slot_records
    )
    if not (
        record[
            "relation_projection"
        ][
            "relation_aware_review_required"
        ]
    )
]


# =========================================================
# Preserve total Evidence lineage
# =========================================================

component_evidence_union = {
    evidence_id
    for record in (
        provisional_seed_slot_records
    )
    for evidence_id in (
        record[
            "component_profile"
        ][
            "component_evidence_ids"
        ]
    )
}


assert (
    component_evidence_union
    == registry_evidence_ids
)


assert (
    len(
        component_evidence_union
    )
    == 4125
)


# =========================================================
# Serialize provisional Seed Pool
# =========================================================

provisional_seed_pool_text = "".join(
    json.dumps(
        record,
        ensure_ascii=False,
        sort_keys=True,
    )
    + "\n"
    for record in (
        provisional_seed_slot_records
    )
)


provisional_seed_pool_sha256 = (
    hashlib.sha256(
        provisional_seed_pool_text.encode(
            "utf-8"
        )
    ).hexdigest()
)


provisional_seed_pool_write_status = (
    publish_frozen_text(
        provisional_seed_pool_path,
        provisional_seed_pool_text,
    )
)


# =========================================================
# Independent disk validation
# =========================================================

disk_provisional_seed_pool = (
    load_jsonl_records(
        provisional_seed_pool_path
    )
)


assert (
    disk_provisional_seed_pool
    == provisional_seed_slot_records
)


assert (
    sha256_file(
        provisional_seed_pool_path
    )
    == provisional_seed_pool_sha256
)


assert (
    len(
        disk_provisional_seed_pool
    )
    == 3817
)


# =========================================================
# Gate
# =========================================================

provisional_seed_slot_status = (
    "PASS"
)


semantic_seed_suitability_review_readiness = (
    "READY_FOR_SEMANTIC_SEED_SUITABILITY_REVIEW_DESIGN"
)


# =========================================================
# Result
# =========================================================

print(
    "=== NB12j PROVISIONAL SEED SLOT CONSTRUCTION ==="
)

print()

print(
    "Provisional Seed Slots:",
    len(
        provisional_seed_slot_records
    ),
)

print(
    "Singleton Slots:",
    len(
        singleton_slots
    ),
)

print(
    "Semantic-Duplicate Component Slots:",
    len(
        duplicate_slots
    ),
)

print(
    "Canonical Claims represented:",
    all_member_claim_occurrences,
)

print(
    "Nonrepresentative Duplicate members preserved:",
    nonrepresentative_member_count,
)


print()

print(
    "Mixed-authority Duplicate Slots:",
    len(
        mixed_authority_duplicate_slots
    ),
)

print(
    "Mixed-authority Slots with Primary representative:",
    sum(
        1
        for slot in (
            mixed_authority_duplicate_slots
        )
        if (
            slot[
                "representative_claim"
            ][
                "authority_class"
            ]
            == "CORE_PRIMARY_KNOWLEDGE"
        )
    ),
)


print()

print(
    "Representative Authority counts:",
    dict(
        sorted(
            representative_authority_counts.items()
        )
    ),
)

print(
    "Representative Source-Type counts:",
    dict(
        sorted(
            representative_source_type_counts.items()
        )
    ),
)

print(
    "Representative Claim-Type counts:",
    dict(
        sorted(
            representative_claim_type_counts.items()
        )
    ),
)


print()

print(
    "=== RELATION PROJECTION ==="
)

print(
    "Internal projected Relations:",
    dict(
        sorted(
            internal_projected_relation_counts.items()
        )
    ),
)

print(
    "External projected Relations:",
    dict(
        sorted(
            external_projected_relation_counts.items()
        )
    ),
)

print(
    "External Semantic-Duplicate Relations:",
    external_projected_relation_counts[
        "SEMANTIC_DUPLICATE"
    ],
)

print(
    "Seed Slots requiring SUBSUMES/PARTIAL review:",
    len(
        relation_aware_seed_slots
    ),
)

print(
    "Seed Slots without external SUBSUMES/PARTIAL review:",
    len(
        relation_isolated_seed_slots
    ),
)


print()

print(
    "Seed Slot review flag counts:",
    dict(
        sorted(
            slot_review_flag_counts.items()
        )
    ),
)


print()

print(
    "Unique Registry Evidence IDs preserved through components:",
    len(
        component_evidence_union
    ),
)


print()

print(
    "Final Seed Slots selected:",
    sum(
        1
        for record in (
            provisional_seed_slot_records
        )
        if (
            record[
                "final_seed_selected"
            ]
        )
    ),
)

print(
    "Question generation started:",
    any(
        record[
            "question_generation_started"
        ]
        for record in (
            provisional_seed_slot_records
        )
    ),
)


print()

print(
    "Provisional Seed Pool:",
    provisional_seed_pool_path.relative_to(
        project_root
    ),
)

print(
    "Provisional Seed Pool SHA-256:",
    provisional_seed_pool_sha256,
)

print(
    "Write status:",
    provisional_seed_pool_write_status,
)


print()

print(
    "API calls performed:",
    0,
)

print(
    "Next stage:",
    "SEMANTIC_SEED_SUITABILITY_REVIEW_DESIGN",
)

print(
    "Readiness:",
    semantic_seed_suitability_review_readiness,
)


print()

print(
    "CODE CELL 11 GATE:",
    provisional_seed_slot_status,
)
````

````python
# Code Cell 13

assert (
    semantic_review_contract_status
    == "PASS"
), (
    "Code Cell 12 must PASS before "
    "Review Packet and Pilot design."
)

assert (
    semantic_review_packet_design_readiness
    == "READY_FOR_REVIEW_PACKET_AND_STRATIFIED_PILOT_DESIGN"
)


# =========================================================
# Configuration
# =========================================================

REVIEW_PACKET_VERSION = (
    "golden_v2_seed_suitability_review_packet_v1"
)


PILOT_SELECTION_VERSION = (
    "golden_v2_seed_suitability_pilot_selection_v1"
)


# =========================================================
# Indexes
# =========================================================

seed_slot_by_id = {
    record[
        "seed_slot_id"
    ]: record
    for record in (
        provisional_seed_slot_records
    )
}


assert (
    len(
        seed_slot_by_id
    )
    == 3817
)


# =========================================================
# Compact Evidence packet
# =========================================================

def make_evidence_packet(
    evidence_id,
):

    evidence = (
        manifest_evidence_by_id[
            evidence_id
        ]
    )


    return {
        "evidence_id": (
            evidence_id
        ),

        "evidence_quote_exact": (
            evidence[
                "evidence_quote_exact"
            ]
        ),

        "locator": (
            evidence[
                "locator"
            ]
        ),

        "source_id": (
            evidence[
                "source_id"
            ]
        ),

        "document_id": (
            evidence[
                "document_id"
            ]
        ),

        "attribution_snapshot": (
            evidence[
                "attribution_snapshot"
            ]
        ),

        "source_type": (
            evidence.get(
                "source_type"
            )
        ),

        "authority_class": (
            evidence.get(
                "authority_class"
            )
        ),

        "claim_discovery_role": (
            evidence.get(
                "claim_discovery_role"
            )
        ),
    }


# =========================================================
# Relation neighbor lookup
# =========================================================

external_relation_context_by_slot = {
    seed_slot_id: []
    for seed_slot_id in (
        seed_slot_by_id
    )
}


for relation in (
    graph_relations
):

    relation_id = (
        relation[
            "relation_id"
        ]
    )

    relation_type = (
        relation[
            "relation_type"
        ]
    )

    claim_id_a = (
        relation[
            "claim_id_a"
        ]
    )

    claim_id_b = (
        relation[
            "claim_id_b"
        ]
    )


    slot_id_a = (
        claim_to_seed_slot_id[
            claim_id_a
        ]
    )

    slot_id_b = (
        claim_to_seed_slot_id[
            claim_id_b
        ]
    )


    # Internal Semantic-Duplicate relations are already
    # represented by Component membership.
    if (
        slot_id_a
        == slot_id_b
    ):

        continue


    # Semantic Duplicate must never remain external.
    assert (
        relation_type
        != "SEMANTIC_DUPLICATE"
    )


    for (
        current_slot_id,
        current_claim_id,
        neighbor_slot_id,
        neighbor_claim_id,
    ) in [
        (
            slot_id_a,
            claim_id_a,
            slot_id_b,
            claim_id_b,
        ),
        (
            slot_id_b,
            claim_id_b,
            slot_id_a,
            claim_id_a,
        ),
    ]:

        neighbor_slot = (
            seed_slot_by_id[
                neighbor_slot_id
            ]
        )


        external_relation_context_by_slot[
            current_slot_id
        ].append({
            "relation_id": (
                relation_id
            ),

            "relation_type": (
                relation_type
            ),

            "claim_id_a": (
                claim_id_a
            ),

            "claim_id_b": (
                claim_id_b
            ),

            "current_related_claim_id": (
                current_claim_id
            ),

            "neighbor_related_claim_id": (
                neighbor_claim_id
            ),

            "neighbor_seed_slot_id": (
                neighbor_slot_id
            ),

            "neighbor_representative_claim_id": (
                neighbor_slot[
                    "representative_claim_id"
                ]
            ),

            "neighbor_representative_claim_text_fa": (
                neighbor_slot[
                    "representative_claim"
                ][
                    "claim_text_fa"
                ]
            ),
        })


for seed_slot_id in (
    external_relation_context_by_slot
):

    external_relation_context_by_slot[
        seed_slot_id
    ] = sorted(
        external_relation_context_by_slot[
            seed_slot_id
        ],
        key=lambda item: (
            item[
                "relation_type"
            ],
            item[
                "relation_id"
            ],
            item[
                "neighbor_seed_slot_id"
            ],
        ),
    )


# =========================================================
# Build Review Packets
# =========================================================

review_packets_by_slot_id = {}


for slot in (
    provisional_seed_slot_records
):

    seed_slot_id = (
        slot[
            "seed_slot_id"
        ]
    )


    representative_claim_id = (
        slot[
            "representative_claim_id"
        ]
    )


    representative_claim = (
        profile_claim_by_id[
            representative_claim_id
        ]
    )


    representative_evidence_packets = [
        make_evidence_packet(
            evidence_id
        )
        for evidence_id in (
            representative_claim[
                "evidence_ids"
            ]
        )
    ]


    component_member_claims = [
        {
            "claim_id": (
                claim_id
            ),

            "claim_text_fa": (
                profile_claim_by_id[
                    claim_id
                ][
                    "claim_text_fa"
                ]
            ),

            "claim_types": (
                profile_claim_by_id[
                    claim_id
                ][
                    "claim_types"
                ]
            ),

            "authority_class": (
                profile_claim_by_id[
                    claim_id
                ][
                    "authority_classes"
                ][
                    0
                ]
            ),

            "source_types": (
                profile_claim_by_id[
                    claim_id
                ][
                    "source_types"
                ]
            ),

            "evidence_ids": (
                profile_claim_by_id[
                    claim_id
                ][
                    "evidence_ids"
                ]
            ),
        }
        for claim_id in (
            slot[
                "member_claim_ids"
            ]
        )
    ]


    review_packet = {
        "review_packet_version": (
            REVIEW_PACKET_VERSION
        ),

        "semantic_review_contract_version": (
            SEMANTIC_REVIEW_CONTRACT_VERSION
        ),

        "semantic_review_contract_sha256": (
            semantic_review_policy_sha256
        ),

        "seed_slot_id": (
            seed_slot_id
        ),

        "slot_kind": (
            slot[
                "slot_kind"
            ]
        ),

        "representative_claim": {
            "claim_id": (
                representative_claim_id
            ),

            "claim_text_fa": (
                representative_claim[
                    "claim_text_fa"
                ]
            ),

            "claim_types": (
                representative_claim[
                    "claim_types"
                ]
            ),

            "authority_class": (
                representative_claim[
                    "authority_classes"
                ][
                    0
                ]
            ),

            "claim_discovery_roles": (
                representative_claim[
                    "claim_discovery_roles"
                ]
            ),

            "source_types": (
                representative_claim[
                    "source_types"
                ]
            ),

            "source_ids": (
                representative_claim[
                    "source_ids"
                ]
            ),

            "source_document_ids": (
                representative_claim[
                    "source_document_ids"
                ]
            ),
        },

        "representative_evidence": (
            representative_evidence_packets
        ),

        "component_context": {
            "member_claim_count": (
                slot[
                    "member_claim_count"
                ]
            ),

            "member_claims": (
                component_member_claims
            ),

            "mixed_authority": (
                slot[
                    "component_profile"
                ][
                    "mixed_authority"
                ]
            ),

            "mixed_source_type": (
                slot[
                    "component_profile"
                ][
                    "mixed_source_type"
                ]
            ),
        },

        "relation_context": (
            external_relation_context_by_slot[
                seed_slot_id
            ]
        ),

        "review_flags": (
            slot[
                "review_flags"
            ]
        ),

        "required_judgment_dimensions": (
            semantic_review_dimensions
        ),

        "allowed_verdicts": sorted(
            semantic_review_verdicts
        ),

        "allowed_reason_codes": sorted(
            semantic_review_reason_codes
        ),
    }


    packet_text = json.dumps(
        review_packet,
        ensure_ascii=False,
        sort_keys=True,
        separators=(
            ",",
            ":",
        ),
    )


    review_packet[
        "review_packet_sha256"
    ] = hashlib.sha256(
        packet_text.encode(
            "utf-8"
        )
    ).hexdigest()


    assert (
        seed_slot_id
        not in review_packets_by_slot_id
    )


    review_packets_by_slot_id[
        seed_slot_id
    ] = review_packet


# =========================================================
# Review Packet validation
# =========================================================

assert (
    len(
        review_packets_by_slot_id
    )
    == 3817
)


assert (
    set(
        review_packets_by_slot_id
    )
    == set(
        seed_slot_by_id
    )
)


assert all(
    packet[
        "representative_evidence"
    ]
    for packet in (
        review_packets_by_slot_id.values()
    )
)


assert all(
    all(
        is_nonempty_value(
            evidence[
                "evidence_quote_exact"
            ]
        )
        for evidence in (
            packet[
                "representative_evidence"
            ]
        )
    )
    for packet in (
        review_packets_by_slot_id.values()
    )
)


assert all(
    all(
        is_nonempty_value(
            evidence[
                "locator"
            ]
        )
        for evidence in (
            packet[
                "representative_evidence"
            ]
        )
    )
    for packet in (
        review_packets_by_slot_id.values()
    )
)


# =========================================================
# Packet size diagnostics
# =========================================================

review_packet_character_lengths = []


for packet in (
    review_packets_by_slot_id.values()
):

    packet_json = json.dumps(
        packet,
        ensure_ascii=False,
        sort_keys=True,
    )


    review_packet_character_lengths.append(
        len(
            packet_json
        )
    )


review_packet_character_profile = (
    numeric_profile(
        review_packet_character_lengths
    )
)


# =========================================================
# Data-driven Pilot Strata
#
# Exact observed combination:
# Authority
# + representative Source Type signature
# + representative Claim Type signature
# + actual Review Risk signature
# =========================================================

def pilot_stratum_key(
    slot,
):

    representative = (
        slot[
            "representative_claim"
        ]
    )


    authority_signature = (
        representative[
            "authority_class"
        ]
    )


    source_type_signature = (
        " + ".join(
            sorted(
                representative[
                    "source_types"
                ]
            )
        )
    )


    claim_type_signature = (
        " + ".join(
            sorted(
                representative[
                    "claim_types"
                ]
            )
        )
    )


    risk_signature = (
        " + ".join(
            sorted(
                slot[
                    "review_flags"
                ]
            )
        )
        if (
            slot[
                "review_flags"
            ]
        )
        else "NO_REVIEW_FLAGS"
    )


    return (
        authority_signature,
        source_type_signature,
        claim_type_signature,
        risk_signature,
    )


pilot_strata = defaultdict(
    list
)


for slot in (
    provisional_seed_slot_records
):

    stratum_key = (
        pilot_stratum_key(
            slot
        )
    )


    pilot_strata[
        stratum_key
    ].append(
        slot[
            "seed_slot_id"
        ]
    )


# =========================================================
# Deterministic challenge-oriented representative
#
# Within each observed stratum:
# 1. more external relation context
# 2. larger Semantic-Duplicate component
# 3. more representative Evidence
# 4. shorter Claim as ambiguity stressor
# 5. seed_slot_id tie-break
# =========================================================

def pilot_candidate_rank(
    seed_slot_id,
):

    slot = (
        seed_slot_by_id[
            seed_slot_id
        ]
    )


    relation_context_count = len(
        external_relation_context_by_slot[
            seed_slot_id
        ]
    )


    member_claim_count = (
        slot[
            "member_claim_count"
        ]
    )


    representative_evidence_count = (
        slot[
            "representative_claim"
        ][
            "evidence_count"
        ]
    )


    claim_text_length = len(
        slot[
            "representative_claim"
        ][
            "claim_text_fa"
        ]
    )


    return (
        -relation_context_count,
        -member_claim_count,
        -representative_evidence_count,
        claim_text_length,
        seed_slot_id,
    )


pilot_seed_slot_ids = []


pilot_selection_records = []


for stratum_key in sorted(
    pilot_strata
):

    stratum_member_slot_ids = sorted(
        pilot_strata[
            stratum_key
        ]
    )


    selected_seed_slot_id = min(
        stratum_member_slot_ids,
        key=pilot_candidate_rank,
    )


    pilot_seed_slot_ids.append(
        selected_seed_slot_id
    )


    pilot_selection_records.append({
        "pilot_selection_version": (
            PILOT_SELECTION_VERSION
        ),

        "authority_signature": (
            stratum_key[
                0
            ]
        ),

        "source_type_signature": (
            stratum_key[
                1
            ]
        ),

        "claim_type_signature": (
            stratum_key[
                2
            ]
        ),

        "risk_signature": (
            stratum_key[
                3
            ]
        ),

        "stratum_population": (
            len(
                stratum_member_slot_ids
            )
        ),

        "selected_seed_slot_id": (
            selected_seed_slot_id
        ),

        "selected_review_packet_sha256": (
            review_packets_by_slot_id[
                selected_seed_slot_id
            ][
                "review_packet_sha256"
            ]
        ),
    })


# =========================================================
# Pilot validation
# =========================================================

observed_pilot_strata_count = len(
    pilot_strata
)


pilot_size = len(
    pilot_seed_slot_ids
)


assert (
    pilot_size
    == observed_pilot_strata_count
)


assert (
    len(
        pilot_seed_slot_ids
    )
    == len(
        set(
            pilot_seed_slot_ids
        )
    )
)


assert all(
    seed_slot_id
    in review_packets_by_slot_id
    for seed_slot_id in (
        pilot_seed_slot_ids
    )
)


# Every observed stratum must be represented exactly once.
selected_stratum_keys = [
    pilot_stratum_key(
        seed_slot_by_id[
            seed_slot_id
        ]
    )
    for seed_slot_id in (
        pilot_seed_slot_ids
    )
]


assert (
    len(
        selected_stratum_keys
    )
    == len(
        set(
            selected_stratum_keys
        )
    )
)


assert (
    set(
        selected_stratum_keys
    )
    == set(
        pilot_strata
    )
)


# =========================================================
# Pilot coverage diagnostics
# =========================================================

pilot_slots = [
    seed_slot_by_id[
        seed_slot_id
    ]
    for seed_slot_id in (
        pilot_seed_slot_ids
    )
]


pilot_authority_counts = Counter(
    slot[
        "representative_claim"
    ][
        "authority_class"
    ]
    for slot in (
        pilot_slots
    )
)


pilot_source_type_counts = Counter(
    source_type
    for slot in (
        pilot_slots
    )
    for source_type in (
        slot[
            "representative_claim"
        ][
            "source_types"
        ]
    )
)


pilot_claim_type_counts = Counter(
    claim_type
    for slot in (
        pilot_slots
    )
    for claim_type in (
        slot[
            "representative_claim"
        ][
            "claim_types"
        ]
    )
)


pilot_review_flag_counts = Counter(
    flag
    for slot in (
        pilot_slots
    )
    for flag in (
        slot[
            "review_flags"
        ]
    )
)


# All observed categories must appear in Pilot.
assert (
    set(
        pilot_authority_counts
    )
    == set(
        representative_authority_counts
    )
)


assert (
    set(
        pilot_source_type_counts
    )
    == set(
        representative_source_type_counts
    )
)


assert (
    set(
        pilot_claim_type_counts
    )
    == set(
        representative_claim_type_counts
    )
)


assert (
    set(
        pilot_review_flag_counts
    )
    == set(
        slot_review_flag_counts
    )
)


# =========================================================
# Full Packet Manifest Hash
# =========================================================

review_packet_hash_manifest = [
    {
        "seed_slot_id": (
            seed_slot_id
        ),

        "review_packet_sha256": (
            review_packets_by_slot_id[
                seed_slot_id
            ][
                "review_packet_sha256"
            ]
        ),
    }
    for seed_slot_id in sorted(
        review_packets_by_slot_id
    )
]


review_packet_hash_manifest_text = (
    json.dumps(
        review_packet_hash_manifest,
        ensure_ascii=False,
        sort_keys=True,
        separators=(
            ",",
            ":",
        ),
    )
)


review_packet_hash_manifest_sha256 = (
    hashlib.sha256(
        review_packet_hash_manifest_text.encode(
            "utf-8"
        )
    ).hexdigest()
)


# =========================================================
# Pilot Selection Hash
# =========================================================

pilot_selection_text = json.dumps(
    pilot_selection_records,
    ensure_ascii=False,
    sort_keys=True,
    separators=(
        ",",
        ":",
    ),
)


pilot_selection_sha256 = (
    hashlib.sha256(
        pilot_selection_text.encode(
            "utf-8"
        )
    ).hexdigest()
)


# =========================================================
# Stage Gate
# =========================================================

review_packet_and_pilot_design_status = (
    "PASS"
)


pilot_runtime_preflight_readiness = (
    "READY_FOR_SEMANTIC_REVIEW_PILOT_RUNTIME_PREFLIGHT"
)


# =========================================================
# Result
# =========================================================

print(
    "=== NB12j REVIEW PACKET / STRATIFIED PILOT DESIGN ==="
)

print()

print(
    "Review Packets built:",
    len(
        review_packets_by_slot_id
    ),
)

print(
    "Review Packet character profile:",
    review_packet_character_profile,
)

print(
    "Review Packet manifest SHA-256:",
    review_packet_hash_manifest_sha256,
)


print()

print(
    "Observed Pilot Strata:",
    observed_pilot_strata_count,
)

print(
    "Pilot Slots selected:",
    pilot_size,
)

print(
    "Pilot fraction of Full Review:",
    round(
        pilot_size
        / semantic_review_population,
        4,
    ),
)


print()

print(
    "Pilot Authority coverage:",
    dict(
        sorted(
            pilot_authority_counts.items()
        )
    ),
)

print(
    "Pilot Source-Type coverage:",
    dict(
        sorted(
            pilot_source_type_counts.items()
        )
    ),
)

print(
    "Pilot Claim-Type coverage:",
    dict(
        sorted(
            pilot_claim_type_counts.items()
        )
    ),
)

print(
    "Pilot Review-Flag coverage:",
    dict(
        sorted(
            pilot_review_flag_counts.items()
        )
    ),
)


print()

print(
    "All observed Authority classes covered:",
    (
        set(
            pilot_authority_counts
        )
        == set(
            representative_authority_counts
        )
    ),
)

print(
    "All observed Source Types covered:",
    (
        set(
            pilot_source_type_counts
        )
        == set(
            representative_source_type_counts
        )
    ),
)

print(
    "All observed Claim Types covered:",
    (
        set(
            pilot_claim_type_counts
        )
        == set(
            representative_claim_type_counts
        )
    ),
)

print(
    "All observed Review Flags covered:",
    (
        set(
            pilot_review_flag_counts
        )
        == set(
            slot_review_flag_counts
        )
    ),
)


print()

print(
    "Pilot Selection SHA-256:",
    pilot_selection_sha256,
)

print(
    "Pilot fixed target imposed:",
    False,
)

print(
    "API calls performed:",
    0,
)

print(
    "Files modified by this Cell:",
    0,
)


print()

print(
    "Next stage:",
    "SEMANTIC_REVIEW_PILOT_RUNTIME_PREFLIGHT",
)

print(
    "Readiness:",
    pilot_runtime_preflight_readiness,
)


print()

print(
    "CODE CELL 13 GATE:",
    review_packet_and_pilot_design_status,
)
````

````python
# code sel 15
# =========================================================
# NB12j — FINAL FULL-3817 SEMANTIC REVIEW EXECUTION
# RESILIENT v3 — PACED PARALLEL + RETRY + RESUME
#
# REAL PAID EXECUTION
#
# Goals:
# - Process the fixed population of exactly 3817 Seeds
# - Reuse every VALID exact checkpoint
# - INVALID checkpoint => retry that Seed, NOT global stop
# - Malformed / incomplete JSON => retry that Seed
# - Validation failure => retry that Seed
# - HTTP 429 => global cooldown + retry
# - Transient API error => retry that Seed
# - One bad Seed NEVER stops the rest of the population
# - Save every paid response immediately
# - Valid checkpoint = completed
# - Invalid checkpoint = still pending on next resume
# - Jupyter interruption => rerun this Cell and resume
# - 4 parallel workers
# - Global pacing >= 2.7 sec between request starts
# - Hard TOTAL estimated review cost cap = 600,000 toman
# =========================================================


from concurrent.futures import (
    ThreadPoolExecutor,
    wait,
    FIRST_COMPLETED,
)
from collections import Counter
from threading import Lock, Event
from pathlib import Path
import hashlib
import json
import os
import re
import time


# =========================================================
# EXECUTION CONFIG
# =========================================================

EXECUTE_FULL_REVIEW = True

MAX_PARALLEL_WORKERS = 4

# ~22.22 request starts/minute maximum.
MIN_SECONDS_BETWEEN_REQUEST_STARTS = 2.7

# Global cooldown after HTTP 429.
RATE_LIMIT_COOLDOWN_SECONDS = 70.0

# ---------------------------------------------------------
# Retry policy
#
# Attempt count means NEW paid API calls during THIS run.
#
# Example:
# MAX_TOTAL_ATTEMPTS_PER_SEED = 3
# => original attempt + up to 2 retries.
#
# Invalid checkpoints from PREVIOUS runs do not consume
# this run's attempt counter. They simply cause the Seed
# to enter the pending queue again.
# ---------------------------------------------------------

MAX_TOTAL_ATTEMPTS_PER_SEED = 3

# 429 may use the same total attempt budget.
MAX_429_RETRIES_PER_SEED = 2

# Backoff for non-429 API exceptions.
API_ERROR_RETRY_BASE_SECONDS = 8.0

# Backoff for malformed/invalid model output.
INVALID_RESPONSE_RETRY_BASE_SECONDS = 2.0

FULL_REVIEW_HARD_COST_CAP_TOMAN = 600_000

PROGRESS_EVERY_N_COMPLETIONS = 25


assert EXECUTE_FULL_REVIEW is True
assert MAX_PARALLEL_WORKERS >= 1
assert MIN_SECONDS_BETWEEN_REQUEST_STARTS > 0
assert MAX_TOTAL_ATTEMPTS_PER_SEED >= 1
assert MAX_429_RETRIES_PER_SEED >= 0


# =========================================================
# REQUIRE EXISTING NB12j CONTROLLER STATE
# =========================================================

required_final_names = [
    "full_seed_slot_ids",
    "FULL_REVIEW_POLICY_VERSION",
    "PROMPT_VERSION",
    "SYSTEM_PROMPT",
    "MAX_TOKENS",
    "TEMPERATURE",
    "TOP_P",
    "MAX_ACCEPTABLE_RESIDUAL_REASONING_TOKENS",
    "semantic_review_client",
    "semantic_review_model",
    "review_packets_by_slot_id",
    "build_request_identity",
    "validate_cached_or_runtime_record",
    "derive_canonical_reason_codes",
    "locate_exact_checkpoint",
    "full_cache_dir",
    "manifest_by_seed_id",
    "input_token_slope",
    "input_token_intercept",
    "p95_positive_input_residual",
    "p95_output_tokens",
    "INPUT_TOMAN_PER_TOKEN",
    "OUTPUT_TOMAN_PER_TOKEN",
]


missing_final_names = [
    name
    for name in required_final_names
    if name not in globals()
]


assert not missing_final_names, (
    "Missing required NB12j objects: "
    + ", ".join(missing_final_names)
)


assert len(full_seed_slot_ids) == 3817
assert len(set(full_seed_slot_ids)) == 3817


full_cache_dir.mkdir(
    parents=True,
    exist_ok=True,
)


# =========================================================
# BASIC UTILITIES
# =========================================================

def estimated_cost_toman(
    input_tokens,
    output_tokens,
):
    return (
        input_tokens * INPUT_TOMAN_PER_TOKEN
        + output_tokens * OUTPUT_TOMAN_PER_TOKEN
    )


def clean_model_json(content):

    if not isinstance(content, str):
        return ""

    clean = content.strip()

    if clean.startswith("```"):

        clean = re.sub(
            r"^```(?:json)?\s*",
            "",
            clean,
            flags=re.IGNORECASE,
        )

        clean = re.sub(
            r"\s*```$",
            "",
            clean,
        )

    return clean.strip()


def is_rate_limit_exception(exc):

    error_text = str(exc).lower()
    error_type = type(exc).__name__.lower()

    return (
        "ratelimit" in error_type
        or "rate_limit" in error_type
        or "429" in error_text
        or "too many requests" in error_text
    )


def is_nonretryable_api_exception(exc):
    """
    Conservative detection of errors where repeating the
    exact request is unlikely to help.

    These do NOT stop the population. They only exhaust
    this Seed immediately.
    """

    text = str(exc).lower()

    markers = [
        "invalid api key",
        "incorrect api key",
        "authentication",
        "unauthorized",
        "forbidden",
        "insufficient balance",
        "insufficient credit",
        "billing",
    ]

    return any(
        marker in text
        for marker in markers
    )


def safe_read_json(path):

    try:
        return json.loads(
            path.read_text(
                encoding="utf-8"
            )
        )

    except Exception:
        return None


def atomic_write_json(
    path,
    payload,
):
    """
    Prevent a partially-written checkpoint from replacing
    a healthy file if the kernel/process dies mid-write.
    """

    path = Path(path)

    temp_path = path.with_name(
        path.name
        + ".tmp"
    )

    temp_path.write_text(
        json.dumps(
            payload,
            ensure_ascii=False,
            sort_keys=True,
            indent=2,
        )
        + "\n",
        encoding="utf-8",
    )

    os.replace(
        temp_path,
        path,
    )


# =========================================================
# CONSERVATIVE PER-REQUEST BUDGET RESERVATION
# =========================================================

def conservative_request_reserve_toman(
    seed_slot_id,
):

    prompt_chars = (
        manifest_by_seed_id[
            seed_slot_id
        ][
            "prompt_character_count"
        ]
    )

    predicted_input = max(
        1,
        round(
            input_token_intercept
            + (
                input_token_slope
                * prompt_chars
            )
            + p95_positive_input_residual
        ),
    )

    predicted_output = int(
        p95_output_tokens
    )

    return estimated_cost_toman(
        predicted_input,
        predicted_output,
    )


# =========================================================
# GLOBAL EXECUTION CONTROL
#
# stop_event is reserved ONLY for genuinely GLOBAL reasons:
# - hard budget ceiling
#
# A bad Seed does NOT set stop_event.
# =========================================================

stop_event = Event()

global_stop_reason_lock = Lock()

global_stop_reason = {
    "reason": None,
}


def request_global_stop(reason):

    with global_stop_reason_lock:

        if (
            global_stop_reason[
                "reason"
            ]
            is None
        ):
            global_stop_reason[
                "reason"
            ] = reason

    stop_event.set()


# =========================================================
# GLOBAL PACING + 429 COOLDOWN
# =========================================================

request_control_lock = Lock()

last_request_start_monotonic = [
    None
]

cooldown_until_monotonic = [
    0.0
]


def activate_global_cooldown(seconds):

    with request_control_lock:

        proposed_until = (
            time.monotonic()
            + seconds
        )

        cooldown_until_monotonic[0] = max(
            cooldown_until_monotonic[0],
            proposed_until,
        )


def acquire_paced_request_start():

    while not stop_event.is_set():

        sleep_seconds = 0.05

        with request_control_lock:

            now = time.monotonic()

            cooldown_remaining = (
                cooldown_until_monotonic[0]
                - now
            )

            if cooldown_remaining > 0:

                sleep_seconds = min(
                    cooldown_remaining,
                    0.5,
                )

            else:

                previous_start = (
                    last_request_start_monotonic[
                        0
                    ]
                )

                if previous_start is None:

                    last_request_start_monotonic[
                        0
                    ] = now

                    return True

                elapsed_since_start = (
                    now
                    - previous_start
                )

                spacing_remaining = (
                    MIN_SECONDS_BETWEEN_REQUEST_STARTS
                    - elapsed_since_start
                )

                if spacing_remaining <= 0:

                    last_request_start_monotonic[
                        0
                    ] = now

                    return True

                sleep_seconds = min(
                    max(
                        spacing_remaining,
                        0.05,
                    ),
                    0.5,
                )

        time.sleep(
            sleep_seconds
        )

    return False


# =========================================================
# SHARED BUDGET STATE
# =========================================================

budget_lock = Lock()

budget_state = {
    "actual_new_input_tokens": 0,
    "actual_new_output_tokens": 0,
    "actual_new_total_tokens": 0,
    "actual_new_cost_toman": 0.0,
    "reserved_toman": 0.0,
}


def try_reserve_budget(
    seed_slot_id,
    cached_cost_toman,
):

    reserve = (
        conservative_request_reserve_toman(
            seed_slot_id
        )
    )

    with budget_lock:

        projected_total_cost = (
            cached_cost_toman
            + budget_state[
                "actual_new_cost_toman"
            ]
            + budget_state[
                "reserved_toman"
            ]
            + reserve
        )

        if (
            projected_total_cost
            > FULL_REVIEW_HARD_COST_CAP_TOMAN
        ):
            return None

        budget_state[
            "reserved_toman"
        ] += reserve

    return reserve


def settle_budget(
    *,
    reserved_toman,
    input_tokens,
    output_tokens,
    total_tokens,
):

    actual_cost = estimated_cost_toman(
        input_tokens,
        output_tokens,
    )

    with budget_lock:

        budget_state[
            "reserved_toman"
        ] -= reserved_toman

        budget_state[
            "actual_new_input_tokens"
        ] += input_tokens

        budget_state[
            "actual_new_output_tokens"
        ] += output_tokens

        budget_state[
            "actual_new_total_tokens"
        ] += total_tokens

        budget_state[
            "actual_new_cost_toman"
        ] += actual_cost

    return actual_cost


def release_budget_reservation(
    reserved_toman,
):

    if reserved_toman is None:
        return

    with budget_lock:

        budget_state[
            "reserved_toman"
        ] -= reserved_toman


# =========================================================
# INITIAL DISK RE-SCAN
#
# IMPORTANT REPAIR:
#
# INVALID checkpoint != fatal.
#
# Valid checkpoint:
#   reuse.
#
# Invalid/malformed checkpoint:
#   put Seed back into pending queue.
# =========================================================

already_valid_results = {}

pending_seed_slot_ids = []

invalid_existing_checkpoints = []


for seed_slot_id in full_seed_slot_ids:

    identity = (
        build_request_identity(
            seed_slot_id
        )
    )

    prompt_sha256 = (
        identity[
            "prompt_sha256"
        ]
    )

    checkpoint_path = (
        locate_exact_checkpoint(
            seed_slot_id,
            prompt_sha256,
        )
    )

    if checkpoint_path is None:

        pending_seed_slot_ids.append(
            seed_slot_id
        )

        continue

    record = safe_read_json(
        checkpoint_path
    )

    # -----------------------------------------------------
    # Unreadable / malformed checkpoint
    # => retry this Seed.
    # -----------------------------------------------------

    if not isinstance(record, dict):

        invalid_existing_checkpoints.append({
            "seed_slot_id": seed_slot_id,
            "checkpoint_path": str(
                checkpoint_path
            ),
            "reason": (
                "CHECKPOINT_JSON_UNREADABLE"
            ),
        })

        pending_seed_slot_ids.append(
            seed_slot_id
        )

        continue

    # -----------------------------------------------------
    # Wrong identity
    # => do not trust it.
    # -----------------------------------------------------

    if (
        record.get(
            "prompt_sha256"
        )
        != prompt_sha256
    ):

        invalid_existing_checkpoints.append({
            "seed_slot_id": seed_slot_id,
            "checkpoint_path": str(
                checkpoint_path
            ),
            "reason": (
                "PROMPT_SHA256_MISMATCH"
            ),
        })

        pending_seed_slot_ids.append(
            seed_slot_id
        )

        continue

    if (
        record.get(
            "seed_slot_id"
        )
        != seed_slot_id
    ):

        invalid_existing_checkpoints.append({
            "seed_slot_id": seed_slot_id,
            "checkpoint_path": str(
                checkpoint_path
            ),
            "reason": (
                "SEED_SLOT_ID_MISMATCH"
            ),
        })

        pending_seed_slot_ids.append(
            seed_slot_id
        )

        continue

    try:

        (
            checks,
            status,
        ) = (
            validate_cached_or_runtime_record(
                seed_slot_id,
                record,
            )
        )

    except Exception as exc:

        checks = {}

        status = "FAIL"

        invalid_existing_checkpoints.append({
            "seed_slot_id": seed_slot_id,
            "checkpoint_path": str(
                checkpoint_path
            ),
            "reason": (
                "VALIDATOR_EXCEPTION"
            ),
            "error": (
                f"{type(exc).__name__}: {exc}"
            ),
        })

        pending_seed_slot_ids.append(
            seed_slot_id
        )

        continue

    if status != "PASS":

        failed_checks = [
            name
            for name, value
            in checks.items()
            if not value
        ]

        invalid_existing_checkpoints.append({
            "seed_slot_id": seed_slot_id,
            "checkpoint_path": str(
                checkpoint_path
            ),
            "reason": (
                "CHECKPOINT_VALIDATION_FAIL"
            ),
            "failed_checks": (
                failed_checks
            ),
        })

        pending_seed_slot_ids.append(
            seed_slot_id
        )

        continue

    record = dict(record)

    record[
        "current_validation_checks"
    ] = checks

    record[
        "current_validation_status"
    ] = status

    already_valid_results[
        seed_slot_id
    ] = record


# =========================================================
# POPULATION INTEGRITY
# =========================================================

assert (
    len(already_valid_results)
    + len(pending_seed_slot_ids)
    == 3817
), (
    "Initial valid + pending population must equal 3817."
)


assert (
    set(already_valid_results)
    .isdisjoint(
        set(pending_seed_slot_ids)
    )
)


# =========================================================
# CURRENT VALID CACHED COST
# =========================================================

cached_input_tokens = sum(
    int(
        (
            record.get(
                "usage"
            )
            or {}
        ).get(
            "input_tokens",
            0,
        )
        or 0
    )
    for record in (
        already_valid_results.values()
    )
)


cached_output_tokens = sum(
    int(
        (
            record.get(
                "usage"
            )
            or {}
        ).get(
            "output_tokens",
            0,
        )
        or 0
    )
    for record in (
        already_valid_results.values()
    )
)


current_cached_cost_toman = (
    estimated_cost_toman(
        cached_input_tokens,
        cached_output_tokens,
    )
)


# =========================================================
# PREFLIGHT
# =========================================================

print(
    "=== NB12j RESILIENT FULL REVIEW PREFLIGHT ==="
)

print()

print(
    "Total population:",
    len(full_seed_slot_ids),
)

print(
    "Valid cached reviews:",
    len(already_valid_results),
)

print(
    "Pending Seeds:",
    len(pending_seed_slot_ids),
)

print(
    "Invalid old checkpoints queued for retry:",
    len(invalid_existing_checkpoints),
)

print(
    "Parallel workers:",
    MAX_PARALLEL_WORKERS,
)

print(
    "Minimum seconds between request starts:",
    MIN_SECONDS_BETWEEN_REQUEST_STARTS,
)

print(
    "Maximum theoretical request starts/minute:",
    round(
        60
        / MIN_SECONDS_BETWEEN_REQUEST_STARTS,
        2,
    ),
)

print(
    "Maximum paid attempts per Seed this run:",
    MAX_TOTAL_ATTEMPTS_PER_SEED,
)

print(
    "429 cooldown:",
    RATE_LIMIT_COOLDOWN_SECONDS,
    "seconds",
)

print(
    "Current VALID cached estimated cost:",
    round(
        current_cached_cost_toman
    ),
    "toman",
)

print(
    "Hard TOTAL estimated cost cap:",
    FULL_REVIEW_HARD_COST_CAP_TOMAN,
    "toman",
)


if invalid_existing_checkpoints:

    print()

    print(
        "First invalid old checkpoints:"
    )

    for item in (
        invalid_existing_checkpoints[
            :10
        ]
    ):

        print(
            " -",
            item,
        )


print()


# =========================================================
# SINGLE SEED EXECUTION
# =========================================================

def execute_one_seed(
    seed_slot_id,
):

    if stop_event.is_set():

        return {
            "seed_slot_id": seed_slot_id,
            "execution_status": (
                "SKIPPED_GLOBAL_STOP"
            ),
        }

    identity = (
        build_request_identity(
            seed_slot_id
        )
    )

    packet = (
        identity[
            "packet"
        ]
    )

    user_prompt = (
        identity[
            "user_prompt"
        ]
    )

    prompt_sha256 = (
        identity[
            "prompt_sha256"
        ]
    )

    checkpoint_filename = (
        f"{seed_slot_id}"
        f"__{prompt_sha256[:16]}"
        ".json"
    )

    checkpoint_path = (
        full_cache_dir
        / checkpoint_filename
    )

    # =====================================================
    # RACE-SAFE VALID CACHE CHECK
    # =====================================================

    existing_path = (
        locate_exact_checkpoint(
            seed_slot_id,
            prompt_sha256,
        )
    )

    if existing_path is not None:

        existing_record = (
            safe_read_json(
                existing_path
            )
        )

        if isinstance(
            existing_record,
            dict,
        ):

            try:

                (
                    checks,
                    status,
                ) = (
                    validate_cached_or_runtime_record(
                        seed_slot_id,
                        existing_record,
                    )
                )

            except Exception:

                status = "FAIL"

            if status == "PASS":

                return {
                    "seed_slot_id": (
                        seed_slot_id
                    ),
                    "execution_status": (
                        "CACHE"
                    ),
                    "record": (
                        existing_record
                    ),
                }

    # =====================================================
    # ATTEMPT LOOP
    # =====================================================

    attempt_history = []

    rate_limit_retry_count = 0

    total_paid_attempts = 0

    while (
        total_paid_attempts
        < MAX_TOTAL_ATTEMPTS_PER_SEED
    ):

        if stop_event.is_set():

            return {
                "seed_slot_id": (
                    seed_slot_id
                ),
                "execution_status": (
                    "SKIPPED_GLOBAL_STOP"
                ),
                "attempt_history": (
                    attempt_history
                ),
            }

        # -------------------------------------------------
        # Reserve budget separately for EACH paid call.
        # -------------------------------------------------

        reserved_toman = (
            try_reserve_budget(
                seed_slot_id,
                current_cached_cost_toman,
            )
        )

        if reserved_toman is None:

            request_global_stop(
                "HARD_COST_CAP_REACHED"
            )

            return {
                "seed_slot_id": (
                    seed_slot_id
                ),
                "execution_status": (
                    "BUDGET_CAP_BLOCKED"
                ),
                "attempt_history": (
                    attempt_history
                ),
            }

        # -------------------------------------------------
        # Global pacing.
        # -------------------------------------------------

        if not acquire_paced_request_start():

            release_budget_reservation(
                reserved_toman
            )

            return {
                "seed_slot_id": (
                    seed_slot_id
                ),
                "execution_status": (
                    "SKIPPED_GLOBAL_STOP"
                ),
                "attempt_history": (
                    attempt_history
                ),
            }

        total_paid_attempts += 1

        attempt_number = (
            total_paid_attempts
        )

        started_at = time.time()

        # -------------------------------------------------
        # API CALL
        # -------------------------------------------------

        try:

            response = (
                semantic_review_client
                .chat
                .completions
                .create(
                    model=(
                        semantic_review_model
                    ),

                    messages=[
                        {
                            "role": "system",
                            "content": (
                                SYSTEM_PROMPT
                            ),
                        },
                        {
                            "role": "user",
                            "content": (
                                user_prompt
                            ),
                        },
                    ],

                    temperature=(
                        TEMPERATURE
                    ),

                    top_p=(
                        TOP_P
                    ),

                    max_tokens=(
                        MAX_TOKENS
                    ),

                    extra_body={
                        "enable_thinking": False,
                    },
                )
            )

        except Exception as exc:

            release_budget_reservation(
                reserved_toman
            )

            error_text = (
                f"{type(exc).__name__}: {exc}"
            )

            # ---------------------------------------------
            # HTTP 429
            # ---------------------------------------------

            if is_rate_limit_exception(
                exc
            ):

                rate_limit_retry_count += 1

                attempt_history.append({
                    "attempt": (
                        attempt_number
                    ),
                    "status": (
                        "HTTP_429"
                    ),
                    "error": (
                        error_text
                    ),
                })

                if (
                    rate_limit_retry_count
                    <= MAX_429_RETRIES_PER_SEED
                    and total_paid_attempts
                    < MAX_TOTAL_ATTEMPTS_PER_SEED
                ):

                    activate_global_cooldown(
                        RATE_LIMIT_COOLDOWN_SECONDS
                    )

                    print(
                        "[429]",
                        seed_slot_id,
                        "| attempt",
                        attempt_number,
                        "| global cooldown",
                        int(
                            RATE_LIMIT_COOLDOWN_SECONDS
                        ),
                        "sec",
                        "| continuing population",
                        flush=True,
                    )

                    continue

                return {
                    "seed_slot_id": (
                        seed_slot_id
                    ),
                    "execution_status": (
                        "SEED_RETRY_EXHAUSTED"
                    ),
                    "failure_kind": (
                        "HTTP_429"
                    ),
                    "error": (
                        error_text
                    ),
                    "attempt_history": (
                        attempt_history
                    ),
                }

            # ---------------------------------------------
            # Authentication/billing/etc.
            #
            # Do not waste retries on THIS Seed.
            # Population scheduler still remains alive.
            # ---------------------------------------------

            if is_nonretryable_api_exception(
                exc
            ):

                attempt_history.append({
                    "attempt": (
                        attempt_number
                    ),
                    "status": (
                        "NONRETRYABLE_API_ERROR"
                    ),
                    "error": (
                        error_text
                    ),
                })

                return {
                    "seed_slot_id": (
                        seed_slot_id
                    ),
                    "execution_status": (
                        "SEED_RETRY_EXHAUSTED"
                    ),
                    "failure_kind": (
                        "NONRETRYABLE_API_ERROR"
                    ),
                    "error": (
                        error_text
                    ),
                    "attempt_history": (
                        attempt_history
                    ),
                }

            # ---------------------------------------------
            # Other API error.
            # Retry Seed without stopping population.
            # ---------------------------------------------

            attempt_history.append({
                "attempt": (
                    attempt_number
                ),
                "status": (
                    "API_ERROR"
                ),
                "error": (
                    error_text
                ),
            })

            if (
                total_paid_attempts
                < MAX_TOTAL_ATTEMPTS_PER_SEED
            ):

                backoff = (
                    API_ERROR_RETRY_BASE_SECONDS
                    * attempt_number
                )

                print(
                    "[API RETRY]",
                    seed_slot_id,
                    "| attempt",
                    attempt_number,
                    "|",
                    error_text,
                    "| retry in",
                    round(
                        backoff,
                        1,
                    ),
                    "sec",
                    flush=True,
                )

                time.sleep(
                    backoff
                )

                continue

            return {
                "seed_slot_id": (
                    seed_slot_id
                ),
                "execution_status": (
                    "SEED_RETRY_EXHAUSTED"
                ),
                "failure_kind": (
                    "API_ERROR"
                ),
                "error": (
                    error_text
                ),
                "attempt_history": (
                    attempt_history
                ),
            }

        elapsed_seconds = (
            time.time()
            - started_at
        )

        # =================================================
        # RESPONSE EXTRACTION
        # =================================================

        try:

            choice = response.choices[0]

            message = choice.message

            finish_reason = getattr(
                choice,
                "finish_reason",
                None,
            )

            content = getattr(
                message,
                "content",
                None,
            )

            usage = getattr(
                response,
                "usage",
                None,
            )

        except Exception as exc:

            release_budget_reservation(
                reserved_toman
            )

            error_text = (
                f"{type(exc).__name__}: {exc}"
            )

            attempt_history.append({
                "attempt": (
                    attempt_number
                ),
                "status": (
                    "RESPONSE_EXTRACTION_ERROR"
                ),
                "error": (
                    error_text
                ),
            })

            if (
                total_paid_attempts
                < MAX_TOTAL_ATTEMPTS_PER_SEED
            ):

                time.sleep(
                    INVALID_RESPONSE_RETRY_BASE_SECONDS
                    * attempt_number
                )

                continue

            return {
                "seed_slot_id": (
                    seed_slot_id
                ),
                "execution_status": (
                    "SEED_RETRY_EXHAUSTED"
                ),
                "failure_kind": (
                    "RESPONSE_EXTRACTION_ERROR"
                ),
                "error": (
                    error_text
                ),
                "attempt_history": (
                    attempt_history
                ),
            }

        # =================================================
        # USAGE
        # =================================================

        if usage is None:

            # We do not know exact provider usage.
            # Release reservation because there is no
            # observed usage object to settle against.
            release_budget_reservation(
                reserved_toman
            )

            attempt_history.append({
                "attempt": (
                    attempt_number
                ),
                "status": (
                    "MISSING_USAGE"
                ),
            })

            if (
                total_paid_attempts
                < MAX_TOTAL_ATTEMPTS_PER_SEED
            ):

                time.sleep(
                    INVALID_RESPONSE_RETRY_BASE_SECONDS
                    * attempt_number
                )

                continue

            return {
                "seed_slot_id": (
                    seed_slot_id
                ),
                "execution_status": (
                    "SEED_RETRY_EXHAUSTED"
                ),
                "failure_kind": (
                    "MISSING_USAGE"
                ),
                "attempt_history": (
                    attempt_history
                ),
            }

        input_tokens = int(
            getattr(
                usage,
                "prompt_tokens",
                0,
            )
            or 0
        )

        output_tokens = int(
            getattr(
                usage,
                "completion_tokens",
                0,
            )
            or 0
        )

        total_tokens = int(
            getattr(
                usage,
                "total_tokens",
                0,
            )
            or (
                input_tokens
                + output_tokens
            )
        )

        completion_details = getattr(
            usage,
            "completion_tokens_details",
            None,
        )

        reasoning_tokens = 0

        if completion_details is not None:

            reported_reasoning = getattr(
                completion_details,
                "reasoning_tokens",
                None,
            )

            if reported_reasoning is not None:

                reasoning_tokens = int(
                    reported_reasoning
                )

        # =================================================
        # JSON PARSE
        # =================================================

        clean_content = (
            clean_model_json(
                content
            )
        )

        parsed_json = None

        parse_error = None

        if clean_content:

            try:

                parsed_json = json.loads(
                    clean_content
                )

            except Exception as exc:

                parse_error = (
                    f"{type(exc).__name__}: {exc}"
                )

        else:

            parse_error = (
                "NO_FINAL_CONTENT"
            )

        # =================================================
        # SETTLE OBSERVED COST
        # =================================================

        actual_cost_toman = (
            settle_budget(
                reserved_toman=(
                    reserved_toman
                ),
                input_tokens=(
                    input_tokens
                ),
                output_tokens=(
                    output_tokens
                ),
                total_tokens=(
                    total_tokens
                ),
            )
        )

        # =================================================
        # SAVE PAID RESPONSE IMMEDIATELY
        # =================================================

        record = {
            "checkpoint_kind": (
                "FULL_SEMANTIC_REVIEW_V1"
            ),

            "policy_version": (
                FULL_REVIEW_POLICY_VERSION
            ),

            "prompt_version": (
                PROMPT_VERSION
            ),

            "prompt_sha256": (
                prompt_sha256
            ),

            "seed_slot_id": (
                seed_slot_id
            ),

            "model": (
                semantic_review_model
            ),

            "no_think": True,

            "attempt_number_this_run": (
                attempt_number
            ),

            "429_retry_count_this_run": (
                rate_limit_retry_count
            ),

            # Compatibility with previous aggregate code.
            "429_retry_count": (
                rate_limit_retry_count
            ),

            "finish_reason": (
                finish_reason
            ),

            "elapsed_seconds": (
                round(
                    elapsed_seconds,
                    4,
                )
            ),

            "usage": {
                "input_tokens": (
                    input_tokens
                ),
                "output_tokens": (
                    output_tokens
                ),
                "total_tokens": (
                    total_tokens
                ),
                "reasoning_tokens": (
                    reasoning_tokens
                ),
            },

            "estimated_cost_toman": (
                actual_cost_toman
            ),

            "raw_content": (
                content
            ),

            "parsed_json": (
                parsed_json
            ),

            "parse_error": (
                parse_error
            ),

            "attempt_history_before_this_response": (
                list(
                    attempt_history
                )
            ),
        }

        # Save BEFORE validation.
        atomic_write_json(
            checkpoint_path,
            record,
        )

        # =================================================
        # VALIDATION
        # =================================================

        try:

            (
                validation_checks,
                validation_status,
            ) = (
                validate_cached_or_runtime_record(
                    seed_slot_id,
                    record,
                )
            )

        except Exception as exc:

            validation_checks = {
                "validator_exception": (
                    False
                )
            }

            validation_status = "FAIL"

            record[
                "validator_exception"
            ] = (
                f"{type(exc).__name__}: {exc}"
            )

        record[
            "current_validation_checks"
        ] = validation_checks

        record[
            "current_validation_status"
        ] = validation_status

        # =================================================
        # PASS
        # =================================================

        if validation_status == "PASS":

            record[
                "canonical_reason_codes"
            ] = (
                derive_canonical_reason_codes(
                    dimensions=(
                        parsed_json[
                            "dimensions"
                        ]
                    ),
                    review_flags=(
                        packet[
                            "review_flags"
                        ]
                    ),
                )
            )

            record[
                "reason_code_source"
            ] = (
                "DETERMINISTIC_LOCAL_DERIVATION"
            )

            record[
                "attempt_history"
            ] = (
                attempt_history
                + [{
                    "attempt": (
                        attempt_number
                    ),
                    "status": (
                        "PASS"
                    ),
                }]
            )

            atomic_write_json(
                checkpoint_path,
                record,
            )

            return {
                "seed_slot_id": (
                    seed_slot_id
                ),
                "execution_status": (
                    "PASS"
                ),
                "record": (
                    record
                ),
                "paid_attempts": (
                    total_paid_attempts
                ),
            }

        # =================================================
        # INVALID MODEL RESPONSE
        #
        # CRITICAL REPAIR:
        # DO NOT SET GLOBAL stop_event.
        # Retry THIS Seed only.
        # =================================================

        failed_checks = [
            name
            for name, value
            in validation_checks.items()
            if not value
        ]

        invalid_attempt = {
            "attempt": (
                attempt_number
            ),
            "status": (
                "INVALID_RESPONSE"
            ),
            "parse_error": (
                parse_error
            ),
            "finish_reason": (
                finish_reason
            ),
            "failed_checks": (
                failed_checks
            ),
        }

        attempt_history.append(
            invalid_attempt
        )

        record[
            "attempt_history"
        ] = list(
            attempt_history
        )

        atomic_write_json(
            checkpoint_path,
            record,
        )

        if (
            total_paid_attempts
            < MAX_TOTAL_ATTEMPTS_PER_SEED
        ):

            retry_sleep = (
                INVALID_RESPONSE_RETRY_BASE_SECONDS
                * attempt_number
            )

            print(
                "[INVALID RETRY]",
                seed_slot_id,
                "| attempt",
                attempt_number,
                "| parse:",
                parse_error,
                "| failed:",
                failed_checks,
                "| retry in",
                round(
                    retry_sleep,
                    1,
                ),
                "sec",
                flush=True,
            )

            time.sleep(
                retry_sleep
            )

            continue

        # -------------------------------------------------
        # This Seed exhausted retries.
        #
        # IMPORTANT:
        # Population continues.
        #
        # Its last invalid checkpoint remains on disk,
        # therefore rerunning this Cell later automatically
        # queues it again.
        # -------------------------------------------------

        return {
            "seed_slot_id": (
                seed_slot_id
            ),
            "execution_status": (
                "SEED_RETRY_EXHAUSTED"
            ),
            "failure_kind": (
                "INVALID_RESPONSE"
            ),
            "failed_checks": (
                failed_checks
            ),
            "parse_error": (
                parse_error
            ),
            "record": (
                record
            ),
            "attempt_history": (
                attempt_history
            ),
        }

    # Defensive fallback.
    return {
        "seed_slot_id": (
            seed_slot_id
        ),
        "execution_status": (
            "SEED_RETRY_EXHAUSTED"
        ),
        "failure_kind": (
            "UNKNOWN"
        ),
        "attempt_history": (
            attempt_history
        ),
    }


# =========================================================
# PARALLEL SCHEDULER
#
# CRITICAL REPAIR:
# Seed failure is recorded but DOES NOT stop scheduler.
# =========================================================

full_results = dict(
    already_valid_results
)

execution_failures = []

new_success_count = 0

cache_hit_count = 0

seed_retry_exhausted_count = 0

execution_started_at = time.time()

pending_iterator = iter(
    pending_seed_slot_ids
)

in_flight = {}


with ThreadPoolExecutor(
    max_workers=(
        MAX_PARALLEL_WORKERS
    )
) as executor:

    # -----------------------------------------------------
    # INITIAL FILL
    # -----------------------------------------------------

    for _ in range(
        min(
            MAX_PARALLEL_WORKERS,
            len(
                pending_seed_slot_ids
            ),
        )
    ):

        seed_slot_id = next(
            pending_iterator,
            None,
        )

        if seed_slot_id is None:
            break

        future = executor.submit(
            execute_one_seed,
            seed_slot_id,
        )

        in_flight[
            future
        ] = seed_slot_id

    # -----------------------------------------------------
    # ROLLING SCHEDULER
    # -----------------------------------------------------

    while in_flight:

        done_futures, _ = wait(
            list(
                in_flight.keys()
            ),
            return_when=(
                FIRST_COMPLETED
            ),
        )

        for future in done_futures:

            seed_slot_id = (
                in_flight.pop(
                    future
                )
            )

            try:

                result = (
                    future.result()
                )

            except Exception as exc:

                # -----------------------------------------
                # Worker exception is Seed-local.
                # Population continues.
                # -----------------------------------------

                result = {
                    "seed_slot_id": (
                        seed_slot_id
                    ),
                    "execution_status": (
                        "WORKER_EXCEPTION"
                    ),
                    "error": (
                        f"{type(exc).__name__}: {exc}"
                    ),
                }

            execution_status = (
                result.get(
                    "execution_status"
                )
            )

            if execution_status in {
                "PASS",
                "CACHE",
            }:

                record = result.get(
                    "record"
                )

                if record is not None:

                    full_results[
                        seed_slot_id
                    ] = record

                if execution_status == "PASS":

                    new_success_count += 1

                else:

                    cache_hit_count += 1

            elif execution_status == (
                "SEED_RETRY_EXHAUSTED"
            ):

                seed_retry_exhausted_count += 1

                execution_failures.append(
                    result
                )

                print(
                    "[SEED FAILED — CONTINUING]",
                    seed_slot_id,
                    "| kind:",
                    result.get(
                        "failure_kind"
                    ),
                    "| remaining population continues",
                    flush=True,
                )

            elif execution_status in {
                "BUDGET_CAP_BLOCKED",
                "SKIPPED_GLOBAL_STOP",
            }:

                if execution_status == (
                    "BUDGET_CAP_BLOCKED"
                ):

                    execution_failures.append(
                        result
                    )

            else:

                # API/worker/etc. failure is still
                # Seed-local unless a global budget stop
                # has explicitly been requested.
                execution_failures.append(
                    result
                )

                print(
                    "[SEED ERROR — CONTINUING]",
                    seed_slot_id,
                    "| status:",
                    execution_status,
                    flush=True,
                )

            # =================================================
            # PROGRESS
            # =================================================

            completed_total = len(
                full_results
            )

            if (
                new_success_count > 0
                and (
                    new_success_count
                    % PROGRESS_EVERY_N_COMPLETIONS
                    == 0
                )
            ):

                elapsed = (
                    time.time()
                    - execution_started_at
                )

                with budget_lock:

                    current_new_cost = (
                        budget_state[
                            "actual_new_cost_toman"
                        ]
                    )

                print(
                    f"[{completed_total}/3817]",
                    "| new PASS:",
                    new_success_count,
                    "| seed failures:",
                    seed_retry_exhausted_count,
                    "| remaining valid needed:",
                    (
                        3817
                        - completed_total
                    ),
                    "| elapsed:",
                    round(
                        elapsed / 60,
                        1,
                    ),
                    "min",
                    "| new spend ≈",
                    round(
                        current_new_cost
                    ),
                    "toman",
                    flush=True,
                )

            # =================================================
            # SUBMIT NEXT
            #
            # Continue after Seed-local failures.
            # Stop submission only for GLOBAL stop_event.
            # =================================================

            if not stop_event.is_set():

                next_seed_slot_id = next(
                    pending_iterator,
                    None,
                )

                if next_seed_slot_id is not None:

                    next_future = (
                        executor.submit(
                            execute_one_seed,
                            next_seed_slot_id,
                        )
                    )

                    in_flight[
                        next_future
                    ] = (
                        next_seed_slot_id
                    )


# =========================================================
# FINAL DISK RE-SCAN
#
# Disk is source of truth.
# =========================================================

final_valid_records = {}

final_invalid_records = []

final_unreadable_records = []


for seed_slot_id in full_seed_slot_ids:

    identity = (
        build_request_identity(
            seed_slot_id
        )
    )

    prompt_sha256 = (
        identity[
            "prompt_sha256"
        ]
    )

    checkpoint_path = (
        locate_exact_checkpoint(
            seed_slot_id,
            prompt_sha256,
        )
    )

    if checkpoint_path is None:
        continue

    record = safe_read_json(
        checkpoint_path
    )

    if not isinstance(
        record,
        dict,
    ):

        final_unreadable_records.append({
            "seed_slot_id": (
                seed_slot_id
            ),
            "checkpoint_path": (
                str(
                    checkpoint_path
                )
            ),
        })

        continue

    try:

        (
            checks,
            status,
        ) = (
            validate_cached_or_runtime_record(
                seed_slot_id,
                record,
            )
        )

    except Exception as exc:

        final_invalid_records.append({
            "seed_slot_id": (
                seed_slot_id
            ),
            "checks": {},
            "validator_error": (
                f"{type(exc).__name__}: {exc}"
            ),
        })

        continue

    if status == "PASS":

        final_valid_records[
            seed_slot_id
        ] = record

    else:

        final_invalid_records.append({
            "seed_slot_id": (
                seed_slot_id
            ),
            "checks": (
                checks
            ),
            "parse_error": (
                record.get(
                    "parse_error"
                )
            ),
        })


final_missing_seed_slot_ids = [
    seed_slot_id
    for seed_slot_id in full_seed_slot_ids
    if (
        seed_slot_id
        not in final_valid_records
    )
]


# =========================================================
# AGGREGATES
# =========================================================

final_verdict_counts = Counter(
    record[
        "parsed_json"
    ][
        "verdict"
    ]
    for record in (
        final_valid_records.values()
    )
)


final_reasoning_distribution = Counter(
    int(
        (
            record.get(
                "usage"
            )
            or {}
        ).get(
            "reasoning_tokens",
            0,
        )
        or 0
    )
    for record in (
        final_valid_records.values()
    )
)


final_429_retry_distribution = Counter(
    int(
        record.get(
            "429_retry_count",
            0,
        )
        or 0
    )
    for record in (
        final_valid_records.values()
    )
)


final_input_tokens = sum(
    int(
        (
            record.get(
                "usage"
            )
            or {}
        ).get(
            "input_tokens",
            0,
        )
        or 0
    )
    for record in (
        final_valid_records.values()
    )
)


final_output_tokens = sum(
    int(
        (
            record.get(
                "usage"
            )
            or {}
        ).get(
            "output_tokens",
            0,
        )
        or 0
    )
    for record in (
        final_valid_records.values()
    )
)


final_total_tokens = sum(
    int(
        (
            record.get(
                "usage"
            )
            or {}
        ).get(
            "total_tokens",
            0,
        )
        or 0
    )
    for record in (
        final_valid_records.values()
    )
)


final_estimated_total_cost_toman = (
    estimated_cost_toman(
        final_input_tokens,
        final_output_tokens,
    )
)


execution_elapsed_seconds = (
    time.time()
    - execution_started_at
)


# =========================================================
# FINAL GATE
# =========================================================

full_review_complete = (
    len(final_valid_records) == 3817
    and not final_missing_seed_slot_ids
    and not final_invalid_records
    and not final_unreadable_records
)


if full_review_complete:

    full_review_execution_status = (
        "PASS"
    )

    full_review_next_readiness = (
        "READY_FOR_GOLDEN_V2_POST_REVIEW_ADJUDICATION"
    )

else:

    full_review_execution_status = (
        "INCOMPLETE_RETRYABLE"
    )

    full_review_next_readiness = (
        "RERUN_THIS_CELL_TO_RETRY_REMAINING_INVALID_OR_MISSING_SEEDS"
    )


# =========================================================
# FINAL OUTPUT
# =========================================================

print()

print(
    "=== NB12j FINAL RESILIENT FULL-3817 SEMANTIC REVIEW ==="
)

print()

print(
    "Target population:",
    3817,
)

print(
    "Final valid reviews:",
    len(
        final_valid_records
    ),
)

print(
    "Final missing/invalid reviews:",
    len(
        final_missing_seed_slot_ids
    ),
)

print(
    "Final invalid checkpoints:",
    len(
        final_invalid_records
    ),
)

print(
    "Final unreadable checkpoints:",
    len(
        final_unreadable_records
    ),
)


print()

print(
    "Valid cached at execution start:",
    len(
        already_valid_results
    ),
)

print(
    "Invalid old checkpoints retried:",
    len(
        invalid_existing_checkpoints
    ),
)

print(
    "NEW successful reviews this execution:",
    new_success_count,
)

print(
    "Seed retry exhaustions this execution:",
    seed_retry_exhausted_count,
)

print(
    "Recorded Seed-local failures:",
    len(
        execution_failures
    ),
)


print()

print(
    "Verdict counts:",
    dict(
        sorted(
            final_verdict_counts.items()
        )
    ),
)

print(
    "Residual reasoning distribution:",
    dict(
        sorted(
            final_reasoning_distribution.items()
        )
    ),
)

print(
    "429 retry distribution:",
    dict(
        sorted(
            final_429_retry_distribution.items()
        )
    ),
)


print()

print(
    "VALID checkpoint input tokens:",
    final_input_tokens,
)

print(
    "VALID checkpoint output tokens:",
    final_output_tokens,
)

print(
    "VALID checkpoint total tokens:",
    final_total_tokens,
)

print(
    "Estimated VALID checkpoint review cost:",
    round(
        final_estimated_total_cost_toman
    ),
    "toman",
)


with budget_lock:

    print(
        "Estimated NEW spend this execution:",
        round(
            budget_state[
                "actual_new_cost_toman"
            ]
        ),
        "toman",
    )


print(
    "Hard TOTAL estimated cost cap:",
    FULL_REVIEW_HARD_COST_CAP_TOMAN,
    "toman",
)


print()

print(
    "Execution elapsed minutes:",
    round(
        execution_elapsed_seconds
        / 60,
        2,
    ),
)

print(
    "Execution elapsed hours:",
    round(
        execution_elapsed_seconds
        / 3600,
        2,
    ),
)


print()

print(
    "Parallel worker pool:",
    MAX_PARALLEL_WORKERS,
)

print(
    "Minimum seconds between request starts:",
    MIN_SECONDS_BETWEEN_REQUEST_STARTS,
)

print(
    "Maximum theoretical request starts/minute:",
    round(
        60
        / MIN_SECONDS_BETWEEN_REQUEST_STARTS,
        2,
    ),
)

print(
    "Maximum paid attempts per Seed this run:",
    MAX_TOTAL_ATTEMPTS_PER_SEED,
)

print(
    "429 cooldown seconds:",
    RATE_LIMIT_COOLDOWN_SECONDS,
)

print(
    "Maximum 429 retries per Seed:",
    MAX_429_RETRIES_PER_SEED,
)

print(
    "Checkpoint / Resume:",
    True,
)


print()

print(
    "Global stop triggered:",
    stop_event.is_set(),
)

print(
    "Global stop reason:",
    global_stop_reason[
        "reason"
    ],
)


if execution_failures:

    print()

    print(
        "First Seed-local failures:"
    )

    for failure in (
        execution_failures[
            :10
        ]
    ):

        print(
            " -",
            failure.get(
                "seed_slot_id"
            ),
            "|",
            failure.get(
                "failure_kind",
                failure.get(
                    "execution_status"
                ),
            ),
        )


if final_missing_seed_slot_ids:

    print()

    print(
        "First remaining Seed IDs:",
        final_missing_seed_slot_ids[
            :10
        ],
    )


print()

print(
    "Frozen Golden artifacts modified:",
    False,
)

print(
    "Next readiness:",
    full_review_next_readiness,
)

print(
    "CODE CELL FINAL FULL REVIEW GATE:",
    full_review_execution_status,
)
````

````python
# code sel 16

# =========================================================
# NB12j — POST-REVIEW ADJUDICATION
# STEP 01 — FULL REVIEW DISCOVERY & PREFLIGHT
#
# READ-ONLY
# NO API CALL
# NO ARTIFACT WRITE
# =========================================================

from collections import Counter, defaultdict
import json


# =========================================================
# REQUIRED EXISTING STATE
# =========================================================

required_names = [
    "full_seed_slot_ids",
    "build_request_identity",
    "locate_exact_checkpoint",
    "validate_cached_or_runtime_record",
]


missing_names = [
    name
    for name in required_names
    if name not in globals()
]


assert not missing_names, (
    "Missing required NB12j state: "
    + ", ".join(missing_names)
)


assert len(full_seed_slot_ids) == 3817
assert len(set(full_seed_slot_ids)) == 3817


# =========================================================
# DISCOVER VALID REVIEW RECORDS
# =========================================================

post_review_records_by_seed_id = {}

post_review_missing_seed_ids = []

post_review_invalid_records = []

post_review_unreadable_records = []


for seed_slot_id in full_seed_slot_ids:

    identity = build_request_identity(
        seed_slot_id
    )

    prompt_sha256 = identity[
        "prompt_sha256"
    ]

    checkpoint_path = locate_exact_checkpoint(
        seed_slot_id,
        prompt_sha256,
    )


    # -----------------------------------------------------
    # Missing checkpoint
    # -----------------------------------------------------

    if checkpoint_path is None:

        post_review_missing_seed_ids.append(
            seed_slot_id
        )

        continue


    # -----------------------------------------------------
    # Read checkpoint
    # -----------------------------------------------------

    try:

        record = json.loads(
            checkpoint_path.read_text(
                encoding="utf-8"
            )
        )

    except Exception as exc:

        post_review_unreadable_records.append({
            "seed_slot_id": seed_slot_id,
            "checkpoint_path": str(
                checkpoint_path
            ),
            "error": (
                f"{type(exc).__name__}: {exc}"
            ),
        })

        continue


    # -----------------------------------------------------
    # Validate checkpoint
    # -----------------------------------------------------

    try:

        checks, status = (
            validate_cached_or_runtime_record(
                seed_slot_id,
                record,
            )
        )

    except Exception as exc:

        post_review_invalid_records.append({
            "seed_slot_id": seed_slot_id,
            "checkpoint_path": str(
                checkpoint_path
            ),
            "reason": (
                "VALIDATOR_EXCEPTION"
            ),
            "error": (
                f"{type(exc).__name__}: {exc}"
            ),
        })

        continue


    if status != "PASS":

        failed_checks = [
            name
            for name, value
            in checks.items()
            if not value
        ]

        post_review_invalid_records.append({
            "seed_slot_id": seed_slot_id,
            "checkpoint_path": str(
                checkpoint_path
            ),
            "reason": (
                "VALIDATION_FAIL"
            ),
            "failed_checks": (
                failed_checks
            ),
        })

        continue


    # -----------------------------------------------------
    # Valid record
    # -----------------------------------------------------

    post_review_records_by_seed_id[
        seed_slot_id
    ] = record


# =========================================================
# POPULATION VALIDATION
# =========================================================

valid_review_count = len(
    post_review_records_by_seed_id
)


assert (
    valid_review_count
    + len(post_review_missing_seed_ids)
    + len(post_review_invalid_records)
    + len(post_review_unreadable_records)
    == 3817
), (
    "Review population accounting does not equal 3817."
)


# =========================================================
# VERDICT EXTRACTION
# =========================================================

verdict_to_seed_ids = defaultdict(
    list
)


verdict_extraction_errors = []


for (
    seed_slot_id,
    record,
) in post_review_records_by_seed_id.items():

    parsed_json = record.get(
        "parsed_json"
    )


    if not isinstance(
        parsed_json,
        dict,
    ):

        verdict_extraction_errors.append({
            "seed_slot_id": seed_slot_id,
            "reason": (
                "PARSED_JSON_NOT_OBJECT"
            ),
        })

        continue


    verdict = parsed_json.get(
        "verdict"
    )


    if not isinstance(
        verdict,
        str,
    ):

        verdict_extraction_errors.append({
            "seed_slot_id": seed_slot_id,
            "reason": (
                "VERDICT_MISSING_OR_INVALID"
            ),
        })

        continue


    verdict_to_seed_ids[
        verdict
    ].append(
        seed_slot_id
    )


# =========================================================
# EXPECTED VERDICT CONTRACT
# =========================================================

allowed_verdicts = {
    "ACCEPT_SEED",
    "HOLD_REPAIR",
    "HOLD_RELATION_REVIEW",
    "REJECT_AS_GOLDEN_SEED",
}


observed_verdicts = set(
    verdict_to_seed_ids.keys()
)


unexpected_verdicts = (
    observed_verdicts
    - allowed_verdicts
)


verdict_counts = Counter({
    verdict: len(seed_ids)
    for (
        verdict,
        seed_ids,
    ) in verdict_to_seed_ids.items()
})


total_extracted_verdicts = sum(
    verdict_counts.values()
)


# =========================================================
# DERIVED REVIEW POPULATIONS
# =========================================================

accepted_seed_slot_ids = sorted(
    verdict_to_seed_ids.get(
        "ACCEPT_SEED",
        [],
    )
)


hold_repair_seed_slot_ids = sorted(
    verdict_to_seed_ids.get(
        "HOLD_REPAIR",
        [],
    )
)


hold_relation_review_seed_slot_ids = sorted(
    verdict_to_seed_ids.get(
        "HOLD_RELATION_REVIEW",
        [],
    )
)


rejected_seed_slot_ids = sorted(
    verdict_to_seed_ids.get(
        "REJECT_AS_GOLDEN_SEED",
        [],
    )
)


# =========================================================
# PREFLIGHT GATE
# =========================================================

post_review_preflight_checks = {
    "target_population_is_3817": (
        len(full_seed_slot_ids)
        == 3817
    ),

    "all_3817_reviews_valid": (
        valid_review_count
        == 3817
    ),

    "no_missing_reviews": (
        len(
            post_review_missing_seed_ids
        )
        == 0
    ),

    "no_invalid_reviews": (
        len(
            post_review_invalid_records
        )
        == 0
    ),

    "no_unreadable_reviews": (
        len(
            post_review_unreadable_records
        )
        == 0
    ),

    "no_verdict_extraction_errors": (
        len(
            verdict_extraction_errors
        )
        == 0
    ),

    "no_unexpected_verdicts": (
        len(
            unexpected_verdicts
        )
        == 0
    ),

    "verdict_population_is_3817": (
        total_extracted_verdicts
        == 3817
    ),
}


post_review_preflight_pass = all(
    post_review_preflight_checks.values()
)


# =========================================================
# OUTPUT
# =========================================================

print(
    "=== NB12j POST-REVIEW ADJUDICATION — STEP 01 ==="
)

print()

print(
    "Target Seed population:",
    len(full_seed_slot_ids),
)

print(
    "Valid Semantic Reviews:",
    valid_review_count,
)

print(
    "Missing Reviews:",
    len(
        post_review_missing_seed_ids
    ),
)

print(
    "Invalid Reviews:",
    len(
        post_review_invalid_records
    ),
)

print(
    "Unreadable Reviews:",
    len(
        post_review_unreadable_records
    ),
)

print(
    "Verdict extraction errors:",
    len(
        verdict_extraction_errors
    ),
)


print()

print(
    "Observed verdicts:",
    sorted(
        observed_verdicts
    ),
)

print(
    "Unexpected verdicts:",
    sorted(
        unexpected_verdicts
    ),
)


print()

print(
    "Verdict counts:"
)

for verdict in sorted(
    allowed_verdicts
):

    print(
        " -",
        verdict,
        ":",
        verdict_counts.get(
            verdict,
            0,
        ),
    )


print()

print(
    "Total extracted verdicts:",
    total_extracted_verdicts,
)


print()

print(
    "Post-review populations:"
)

print(
    " - Accepted:",
    len(
        accepted_seed_slot_ids
    ),
)

print(
    " - Hold Repair:",
    len(
        hold_repair_seed_slot_ids
    ),
)

print(
    " - Hold Relation Review:",
    len(
        hold_relation_review_seed_slot_ids
    ),
)

print(
    " - Rejected:",
    len(
        rejected_seed_slot_ids
    ),
)


print()

print(
    "Validation checks:"
)

for (
    check_name,
    check_value,
) in post_review_preflight_checks.items():

    print(
        " -",
        check_name,
        ":",
        "PASS"
        if check_value
        else "FAIL",
    )


print()


if post_review_preflight_pass:

    print(
        "POST-REVIEW PREFLIGHT: PASS"
    )

else:

    print(
        "POST-REVIEW PREFLIGHT: FAIL"
    )
````
