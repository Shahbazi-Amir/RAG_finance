## ۱. هدف و مرز سند

از پیشنهادهای ثبت‌شدهٔ C069 به تصمیم قابل دفاع دربارهٔ مرجع ارزیابی، سپس اصلاح لازمِ بازیابی و تعیین وضعیت پذیرش برسیم؛ بدون تکرار کارهای موفق و بدون ساخت PASS از تغییر معیار.

این سند برنامهٔ پیشنهادی ناظر است؛ تصویب رسمی قرارداد تازه، پذیرش برچسب، مجوز API/محاسبهٔ سنگین یا پذیرش Release نیست. بازبینی استرا ابتدا روش را بررسی می‌کند. از چت جدید خواسته نمی‌شود به ابتدای NB21 یا NB22 بازگردد.

**مسیر اصلی:** بازبینی روش → مصرف بستهٔ موجود و تعیین دامنهٔ مرجع → سنجش روی رتبه‌های ذخیره‌شده → تصمیم دربارهٔ شکاف واقعی → فقط آزمایش موجه بعدی → Gate و تحویل.

**مسیر جانبی مستقل:** دسته‌بندی و بازیابی مکانیکیِ پاسخ‌های Held در NB22B؛ نباید مانع نتیجهٔ اصلی شود و به معنی داوری تازه یا اعتبار معنایی خودکار نیست.

### کد C069 چه چیزی را کنترل می‌کند؟

- هش بایت C068 و سه فایل بستهٔ C069؛ Parse با رد کلید تکراری و مقادیر غیرمتناهی.
- تعداد و یکتایی ۱۲ پرونده و پوشش دقیق ۱۹۷ evidence_id در پیشنهادها.
- اتصال case_id، هش متن passage، وجود نقل‌قول در همان passage، دلیل غیرخالی و Enumهای مجاز.
- حفظ approved_relevance_label=null و approved_for_evaluation=false.
- اتصال گزارش به SHA منبع و تطبیق شمارش‌های relevance/support.
- نصب فقط داده‌های تأییدشده در مسیر محلی و نوشتن/بازخوانی رسید.

**کنترل‌نشده یا احرازنشده:** صحت معنایی رأی؛ صلاحیت مستقل بازبین؛ کامل‌بودن مرجع؛ پذیرش تغییر قرارداد. فایل annotated Parse و Hash می‌شود، ولی کد ارسالی تطبیق ردیف‌به‌ردیف تمام محتوای آن با source/proposals را انجام نمی‌دهد. فهرست پرچم‌های دامنه و تکرار متن از report نقل می‌شود؛ در همین کد دوباره محاسبه نمی‌شود. اگر این دو برای تصمیم بعدی مصرف شوند، کنترل هدفمندشان لازم است؛ نه تکرار همهٔ داوری‌ها.

وجود نقل‌قول دقیق فقط اتصال متن را کنترل می‌کند، نه درستی استدلال. کد به Imports قبلی وابسته است؛ موفقیت اجرای کاربر ثبت شده، ولی اجرای مستقل از کرنل پاک از آن نتیجه نمی‌شود.

## ۴. واقعیت‌های تصمیم‌ساز قبلی؛ تکرار نشوند

- روی ۵۳۳ Validation، سقف تاریخی Top100 حدود ۷۹٫۰۳٪ و Pool ذخیره‌شده حدود ۹۱٫۸۰٪ بود. این‌ها Oracle هستند، نه عملکرد واقعی یا تضمین دستیابی.
- روی ۲۴ سؤال C064، Recall تاریخی baseline=۶۹٫۴۴٪ و BGE کل Pool=۶۹٫۷۹٪؛ nDCG به‌ترتیب حدود ۶۱٫۱۴٪ و ۶۳٫۰۴٪ بود. بنابراین رتبه‌بندی هم مشکل دارد.
- آزمایش‌های Dense/Word/Char/RRF/Linear، Random Forest، LambdaMART، پنجره‌های E5، embedding BGE-M3 و پایلوت‌های BGE/LLM قبلاً انجام شده‌اند. برای دیدن نتیجه، دوباره اجرا نشوند.
- بازتولید عددهای تاریخی تأیید معنایی تمام Gold نیست؛ تغییر تعریف ارتباط و افزایش امتیاز هم بهبود موتور نیست.
- خطای Parity ONNX ثبت شده، اما علت CPU یا کوچک‌بودن اختلاف اثبات نشده؛ حذف Guard یا Z-score مجوز اختلاط امتیاز ناسازگار نیست.
- zeroهای ویژگی‌های NB22F و نقش baseline_present فرضیهٔ تحلیل‌اند؛ علت افت بدون Ablation ثابت نشده.
- ۲۶۱ Held در NB22B ثبت شده‌اند؛ علت عمدتاً نیم‌فاصله یا ی/ک بودنشان اثبات نشده است.
- NB23 پیش‌نویس قرارداد دارد؛ ساخت آن را تکرار نکن. Release پذیرفته نشده؛ ۷ گروه نسخهٔ authoritative حل‌نشده‌اند. Gate تاریخی ۲ PASS، دو FAIL و ۱۲ BLOCKED و مسائل باز محفوظ‌اند.
- Test تاریخی دیده شده؛ از آن برای انتخاب مدل یا تنظیم پارامتر استفاده نشود. یک اجرای تازهٔ همان Test، Holdout مستقل نمی‌سازد.

## ۵. قواعد قرارداد بازبینی

### سه محور جدا

۱. **اتصال منبع/Locator:** کنترل مکانیکی منبع، نسخه، متن و محدوده؛ از روی متنِ بی‌نام همهٔ آن قابل داوری نیست. به ledger مدیریتی متصل شود، بدون افشای رتبه یا Gold به داور معنایی.

۲. **retrieval_relevance:** اطلاعات مستقیم یا مکمل برای پاسخ به سؤال. اشتراک واژه یا موضوع به‌تنهایی کافی نیست. Enum فعلی: RELEVANT / NOT_RELEVANT / UNCERTAIN.

۳. **answer_support:** SUFFICIENT / PARTIAL / CONTEXT_ONLY / NONE / UNCERTAIN. برای پاسخ کامل‌ندادنِ یک قطعه، آن را خودکار نامرتبط نکن؛ اطلاعات مکمل می‌تواند مرتبط باشد.

Enum موجود برای زیبایی تغییر نکند. اگر واقعاً تغییر قرارداد لازم است، نسخه و Adapter صریح با حفظ اصل داده ثبت شود.

### مرجع، مخرج و ابهام

- سؤال‌ها و شواهد مبهم در دفتر پوشش حفظ شوند؛ UNDETERMINED یا UNCERTAIN خودکار صفر نیست.
- حفظ جمعیت ثابت به معنی اجبار همهٔ سؤال‌ها به مخرج قابل‌سنجش نیست. total/accounted و evaluable و unresolved جدا ثبت شوند؛ حذف پنهان و صفرکردن نامعین هر دو ممنوع‌اند.
- متن تکراری با شناسهٔ Chunk متفاوت طبق قرارداد تاریخی شمارش می‌شود. ادغام در کلاس هم‌ارزی تغییر واحد ارزیابی است و نسخهٔ تازه می‌خواهد؛ دو گروه را خودکار حذف نکن.
- ۱۹۷ جفت، تمام Corpus یا تمام Pool هریک از ۱۲ سؤال نیستند؛ C068 اتحاد Gold تاریخی و Top10های منتخب را گردآورده است. بیرون از این مجموعه، «داوری‌نشده» است نه «نامرتبط».
- فرض «فقط آنچه Gold تاریخی است پاسخ معتبر دارد» پذیرفته نشود. انتظار پاسخ راهنماست، نه تنها جواب مجاز.
- قاعدهٔ افزودن شاهد جایگزین، شاهد مکمل و سؤال وابسته به زمینه باید روشن باشد.
- نمونهٔ ۱۲تایی هدفمند و غیرنماینده است؛ ۴۹ پیشنهاد RELEVANT درصد کیفیت Corpus یا کل Golden نیست.

### مرجعیت و پذیرش

نظر استرا، بازبینی یک مدل است؛ حتی اگر بدون رتبه انجام شود خودکار Human Gold یا مرجع مستقل جمعیت نیست. بازبینی روش، داوری معنایی و پذیرش مالک/بازبینِ دارای صلاحیت و دامنه سه تصمیم جدا هستند. تعداد دو بازبین، آستانهٔ توافق ۵٪ یا هر شرط تازهٔ مشابه مصوب نیست؛ از خود اضافه نشود.

دو مسیر مجازِ روشن:
- **مسیر تشخیصی:** استفاده از پیشنهادهای مدل با برچسب model-assisted/provisional، دامنهٔ محدود و بدون ارتقای وضعیت پذیرش.
- **مسیر مرجع قابل‌مصرف رسمی:** قرارداد، منشأ و صلاحیت بازبینی، تصمیم پذیرش و پوشش لازم احراز و نسخه‌دار شوند. «مدل دوم موافق بود» به‌تنهایی این شرط را برآورده نمی‌کند.

اگر مرجع رسمی فعلاً ممکن نیست، همان بسته با نتیجهٔ روشنِ «تحلیل تشخیصی کامل؛ پذیرش رسمی BLOCKED با علت مشخص» بسته شود؛ مرحلهٔ تشخیص بی‌پایان نشود.

## ۷. چهارخانهٔ مقایسه؛ معنی و شرط اعتبار

| مرجع / موتور | baseline ذخیره‌شده | BGE ذخیره‌شده |
| --- | --- | --- |
| historical reference | H_B | H_R |
| reviewed reference در دامنهٔ مجاز | Q_B | Q_R |

- تغییر مرجع برای baseline: Q_B − H_B.
- تغییر مرجع برای BGE: Q_R − H_R.
- اختلاف مدل روی historical: H_R − H_B.
- اختلاف مدل روی reviewed: Q_R − Q_B.
- اختلاف اثرها: (Q_R − Q_B) − (H_R − H_B).

همهٔ این‌ها باید روی دامنهٔ سؤال‌های مشترک و رتبه‌های یکسانِ همان موتور مقایسه شوند؛ تغییر واحد/مخرج ناشی از مرجع نیز گزارش شود. اگر دامنهٔ قابل‌سنجش بین دو مرجع تفاوت دارد، جدول زوجی بدون توضیح ساخته نشود. این اعداد، تجزیهٔ علّی و جمع‌پذیر «درصد ضعف Gold» و «درصد ضعف موتور» در کل جمعیت نیستند. در مرجع ناقص یا صرفاً پیشنهادی، عدد تشخیصی است و Gate تاریخی را بازنویسی نمی‌کند.

## ۸. شرط توقف و جلوگیری از چرخه

- ورودی با هش/هویت متعارض: همان بستهٔ وابسته متوقف و مسیر دقیق گزارش شود.
- صلاحیت یا مرجع احرازنشده: ادعای رسمی متوقف؛ تحلیل تشخیصی محدود با برچسب درست ممکن است.
- پروندهٔ منفرد نامعین: جدا ثبت شود؛ همهٔ کار مستقل متوقف نشود.
- برای هر بسته فقط یک گزارش نهایی و ادامهٔ مشخص؛ دفترهای معتبر موجود مصرف شوند.
- نظر استرا پیش از اجرای مرحلهٔ جدید خوانده شود؛ این سند خودش نتیجهٔ آن بازبینی نیست.
- هیچ API، دانلود مدل، امتیازدهی تازه، training، کاهش آستانه، تغییر Golden، Commit/Push یا تیک خودکار مجاز نشده است.
- نتایج علمی با شمار سل و تعداد فایل سنجیده نشوند؛ درصد پیشرفت بدون وزن و تعریف مصوب گزارش نشود.

## ۹. قالب کار و خروجی هر بسته

کاربر روی مک و VS Code/Jupyter اجرا می‌کند. پاسخ چت اجرایی شامل توضیح کوتاه، **Markdown داخل باکس و یک سل کامل کد داخل باکس همین چت** باشد؛ فایل Python دانلودی جای آن را نگیرد.

هر بسته باید یکجا: ورودی صریح و شناسه/هش → مصرف نتایج قبلی → کار لازم → کنترل و ذخیرهٔ نسخه‌دار → Reload → Fact/Interpretation/Pending → یک گام بعدی داشته باشد. شمارهٔ بعدی سل پس از بررسی نسخهٔ واقعی انتخاب شود؛ صرف شمارهٔ بعدی در این نقشه ایجاد نشده است.

خروجی حداقل:
- اقدام/زیرکار و دامنه؛
- ورودی و Producer/نسخه؛
- شمار پرونده/جفت/درخواست، جدا؛
- تصمیم و محدودیت؛
- کار موفق مصرف‌شده از Cache و کار تازه؛
- کنترل‌های واقعی و رسید خروجی؛
- شرط پایان و علت هر BLOCKED.

````python
# NB22E-v3-C002 — R01: consume pinned existing results.
# Uses centralized imports from C001; no imports in this cell.
# No API, model execution, semantic judgments, metrics or historical edits.

def run_v3_c002():
    required_names = (
        "os", "json", "hashlib", "Path", "Counter", "defaultdict"
    )
    missing_names = [
        name for name in required_names if name not in globals()
    ]
    if missing_names:
        raise RuntimeError(
            "ابتدا سل Import یعنی C001 را اجرا کنید. نام‌های تعریف‌نشده: "
            + ", ".join(missing_names)
        )

    root = Path(os.environ.get(
        "RAG_FINANCE_ROOT",
        "/Users/macbookpro/AMIR_DATA/00_WORKS/01_PROJECTS/"
        "RAG-projects/RAG_finance"
    )).expanduser().resolve()

    old_dir = root / "reports/phase2/quality_improvement/NB22E.lab.v2"
    out_dir = root / "reports/phase2/quality_improvement/NB22E.lab.v3"

    def need(condition, message):
        if not condition:
            raise ValueError(message)

    def unique_pairs(items):
        obj = {}
        for key, value in items:
            need(key not in obj, "DUPLICATE_JSON_KEY: " + key)
            obj[key] = value
        return obj

    def bad_constant(value):
        raise ValueError("NONFINITE_JSON: " + value)

    def parse(raw):
        return json.loads(
            raw.decode("utf-8"),
            object_pairs_hook=unique_pairs,
            parse_constant=bad_constant
        )

    def canonical(value):
        return json.dumps(
            value,
            ensure_ascii=False,
            sort_keys=True,
            separators=(",", ":"),
            allow_nan=False
        ).encode("utf-8")

    def sha(raw):
        return hashlib.sha256(raw).hexdigest()

    receipts = {}

    def read_pinned(label, path, digest, jsonl=False):
        path = path.resolve()
        need(
            path.is_relative_to(root),
            "PATH_OUTSIDE_PROJECT: " + str(path)
        )
        if not path.is_file():
            raise FileNotFoundError(
                "REQUIRED_INPUT_MISSING: " + str(path)
            )

        raw = path.read_bytes()
        actual_sha = sha(raw)
        need(
            actual_sha == digest,
            "HASH_MISMATCH: " + str(path)
            + "\nExpected: " + digest
            + "\nActual: " + actual_sha
        )

        value = (
            [
                parse(line)
                for line in raw.splitlines()
                if line.strip()
            ]
            if jsonl else parse(raw)
        )

        receipts[label] = {
            "path": str(path.relative_to(root)),
            "file_bytes_sha256": actual_sha
        }
        return value

    def escape(key):
        return str(key).replace("~", "~0").replace("/", "~1")

    def walk(value, pointer=""):
        if isinstance(value, dict):
            yield pointer, value
            for key, child in value.items():
                yield from walk(
                    child, pointer + "/" + escape(key)
                )
        elif isinstance(value, list):
            for index, child in enumerate(value):
                yield from walk(
                    child, pointer + "/" + str(index)
                )

    registration_sha = (
        "a968137c6673e96efcbb8f42e6d653b556"
        "1521af9f140553bae2a5cdee3fd29b"
    )
    registration = read_pinned(
        "C069_registration",
        old_dir / (
            "C069.local_review_registration."
            + registration_sha + ".json"
        ),
        registration_sha
    )
    need(
        registration["status"]
        == "VERIFIED_PROPOSALS_REGISTERED_NOT_GOLD_APPROVED",
        "C069_STATUS_CONFLICT"
    )

    package = {}
    for name in ("proposals", "annotated", "report"):
        binding = registration["verified_artifacts"][name]
        package[name] = read_pinned(
            "C069_" + name,
            root / binding["path"],
            binding["sha256"],
            jsonl=(name != "report")
        )

    source_sha = (
        "97ef987a570cae5f6596e4173efe6488"
        "03fbf975fc59e602499f38455f3209ad"
    )
    source = read_pinned(
        "C068_review_sheet",
        old_dir / (
            "C068.blind_review_sheet." + source_sha + ".jsonl"
        ),
        source_sha,
        jsonl=True
    )

    c070 = read_pinned(
        "C070_existing_ledger",
        old_dir / "C070.diagnostic_contract_and_coverage.v1.json",
        "fd4e5429ff1ffa63c0c94b980252a6dd0dfea620f326b1efa7a8641d7649d644"
    )

    latest_sha = (
        "565cd36340044a6c879045db52d6ac8d"
        "f74eea8e1e3823556202b4541a5d1abf"
    )
    latest = read_pinned(
        "C112_existing_ledger",
        old_dir / (
            "C112.review_work_queue." + latest_sha + ".json"
        ),
        latest_sha
    )
    need(
        isinstance(c070, dict) and isinstance(latest, dict),
        "LEDGER_OBJECT_SCHEMA_REQUIRED"
    )

    # Independent hash verification is distinct from ancestry verification.
    # Different sections are recorded, not assumed to be corrupt.
    snapshot_comparison = {
        "identical_sections": [],
        "different_sections": [],
        "sections_absent_from_C112": [],
        "ancestry_verified_by_this_comparison": False,
        "complete_historical_preservation_established": False
    }

    for key, value in c070.items():
        if key not in latest:
            snapshot_comparison[
                "sections_absent_from_C112"
            ].append(key)
        elif canonical(latest[key]) == canonical(value):
            snapshot_comparison["identical_sections"].append(key)
        else:
            snapshot_comparison["different_sections"].append(key)

    for field in (
        "identical_sections",
        "different_sections",
        "sections_absent_from_C112"
    ):
        snapshot_comparison[field].sort()

    need(
        c070["summary"]["queries_accounted"] == 12
        and c070["summary"]["pairs_accounted"] == 197,
        "C070_REPORTED_POPULATION_CONFLICT"
    )

    queue = latest["evidence_bound_review_work_queue_C112"]
    need(
        queue["gap_review_unit_count"] == 137
        and queue["route_case_counts"] == {
            "REUSE_EXISTING_REVIEW_AND_DETERMINE_RUBRIC_NEED": 178,
            "QUALIFY_RUBRIC_AND_REVIEW_EXISTING_GAP_JUDGMENTS": 115,
            "TRACE_PRIOR_REVIEW_WITHOUT_ASSUMING_REVIEW_ABSENT": 10
        },
        "C112_REPORTED_SNAPSHOT_CONFLICT"
    )

    sources = {}
    case_ids = set()
    for case in source:
        cid = case["case_id"]
        need(
            cid not in case_ids,
            "DUPLICATE_SOURCE_CASE: " + cid
        )
        case_ids.add(cid)

        for evidence in case["evidence"]:
            eid = evidence["evidence_id"]
            need(
                eid not in sources,
                "DUPLICATE_SOURCE_EVIDENCE: " + eid
            )
            need(
                isinstance(evidence["passage"], str),
                "SOURCE_PASSAGE_SCHEMA: " + eid
            )
            sources[eid] = {
                "case_id": cid,
                "passage": evidence["passage"]
            }

    proposals = {}
    for proposal in package["proposals"]:
        eid = proposal["evidence_id"]
        need(
            eid not in proposals,
            "DUPLICATE_PROPOSAL: " + eid
        )
        proposals[eid] = proposal

    need(
        len(case_ids) == 12
        and len(sources) == 197
        and set(sources) == set(proposals),
        "SOURCE_PROPOSAL_POPULATION_CONFLICT"
    )

    for eid, proposal in proposals.items():
        original = sources[eid]
        need(
            proposal["case_id"] == original["case_id"]
            and proposal["passage_sha256"]
            == sha(original["passage"].encode("utf-8")),
            "SOURCE_PROPOSAL_BINDING_CONFLICT: " + eid
        )

    # Targeted mechanical checks; no new semantic judgments.
    scope_cases = set()
    for eid, proposal in proposals.items():
        flag = proposal["question_scope_unresolved"]
        need(
            type(flag) is bool,
            "SCOPE_FLAG_SCHEMA: " + eid
        )
        if flag:
            scope_cases.add(proposal["case_id"])

    passage_groups = defaultdict(list)
    for eid, item in sources.items():
        passage_groups[item["passage"]].append(eid)

    duplicate_groups = sorted(
        sorted(ids)
        for ids in passage_groups.values()
        if len(ids) > 1
    )

    report = package["report"]
    reported_scope = report["scope_review_cases"]
    need(
        isinstance(reported_scope, list)
        and all(isinstance(cid, str) for cid in reported_scope)
        and len(reported_scope) == len(set(reported_scope)),
        "REPORTED_SCOPE_SCHEMA"
    )
    scope_match = sorted(scope_cases) == sorted(reported_scope)

    raw_reported_duplicates = report[
        "exact_duplicate_passage_groups"
    ]
    need(
        isinstance(raw_reported_duplicates, list)
        and all(
            isinstance(group, list)
            and len(group) >= 2
            and all(isinstance(eid, str) for eid in group)
            and len(group) == len(set(group))
            for group in raw_reported_duplicates
        ),
        "REPORTED_DUPLICATE_GROUP_SCHEMA"
    )
    reported_duplicates = sorted(
        sorted(group) for group in raw_reported_duplicates
    )
    duplicates_match = duplicate_groups == reported_duplicates

    # Only actual dictionary/list structures are traversed.
    # Unrecognized annotated representations remain unresolved.
    annotated_nodes = defaultdict(list)
    unexpected_annotated_ids = set()

    for pointer, node in walk(package["annotated"]):
        eid = node.get("evidence_id")
        if not isinstance(eid, str):
            continue
        if eid not in sources:
            unexpected_annotated_ids.add(eid)
        annotated_nodes[eid].append((pointer, node))

    annotation_results = []
    for eid, proposal in sorted(proposals.items()):
        proposal_matches = []
        passage_matches = []
        conflicts = []

        for pointer, node in annotated_nodes.get(eid, []):
            if all(key in node for key in proposal):
                if all(
                    canonical(node[key]) == canonical(value)
                    for key, value in proposal.items()
                ):
                    proposal_matches.append(pointer)
                else:
                    conflicts.append(pointer)

            if "passage" in node:
                case_matches = (
                    "case_id" not in node
                    or node["case_id"] == sources[eid]["case_id"]
                )
                if (
                    node["passage"] == sources[eid]["passage"]
                    and case_matches
                ):
                    passage_matches.append(pointer)
                else:
                    conflicts.append(pointer)

        complete = bool(
            proposal_matches
            and passage_matches
            and not conflicts
        )
        annotation_results.append({
            "evidence_id": eid,
            "status": (
                "EXACT_SOURCE_AND_PROPOSAL_CONTENT_BOUND"
                if complete else
                "ANNOTATED_CONTENT_CONFLICT"
                if conflicts else
                "ANNOTATED_REPRESENTATION_NOT_FULLY_BOUND"
            ),
            "proposal_match_pointers": proposal_matches,
            "passage_match_pointers": passage_matches,
            "conflict_pointers": sorted(set(conflicts))
        })

    annotated_complete = (
        not unexpected_annotated_ids
        and all(
            row["status"]
            == "EXACT_SOURCE_AND_PROPOSAL_CONTENT_BOUND"
            for row in annotation_results
        )
    )
    r013_complete = (
        annotated_complete and scope_match and duplicates_match
    )

    pending = []
    if not scope_match:
        pending.append({
            "code": "SCOPE_REPORT_DIFFERENCE",
            "computed": sorted(scope_cases),
            "reported": sorted(reported_scope)
        })
    if not duplicates_match:
        pending.append({
            "code": "DUPLICATE_REPORT_DIFFERENCE",
            "computed": duplicate_groups,
            "reported": reported_duplicates
        })
    if not annotated_complete:
        pending.append({
            "code": "ANNOTATED_BINDING_NOT_COMPLETE",
            "unresolved_or_conflicting_pairs": sum(
                row["status"]
                != "EXACT_SOURCE_AND_PROPOSAL_CONTENT_BOUND"
                for row in annotation_results
            ),
            "unexpected_evidence_ids":
                sorted(unexpected_annotated_ids)
        })

    result = {
        "version": "NB22E_V3_C002_R01_V3",
        "roadmap_action": "R01",
        "status": (
            "R01_COMPLETE_IN_PINNED_SNAPSHOT_SCOPE"
            if r013_complete else
            "R01_INPUTS_CONSUMED_TARGETED_CHECK_REMAINS"
        ),
        "snapshot_scope": "USER_SUPPLIED_V2_THROUGH_C112",
        "latest_unseen_local_or_remote_version_claimed": False,
        "input_receipts": receipts,
        "queries_accounted": len(case_ids),
        "pairs_accounted": len(proposals),
        "C070_C112_snapshot_comparison": snapshot_comparison,
        "checks": {
            "scope_flags_match_report": scope_match,
            "computed_scope_cases": sorted(scope_cases),
            "duplicate_groups_match_report": duplicates_match,
            "computed_duplicate_groups": duplicate_groups,
            "annotated_status_counts": dict(Counter(
                row["status"] for row in annotation_results
            )),
            "unexpected_annotated_evidence_ids":
                sorted(unexpected_annotated_ids)
        },
        "annotated_case_results": annotation_results,
        "pending": pending,
        "subtasks": {
            "R01.1": {
                "status": "DONE_IN_DECLARED_SNAPSHOT_SCOPE",
                "basis": (
                    "C070 and C112 independently hash verified; "
                    "section differences recorded; "
                    "latest unseen version and ancestry not asserted"
                )
            },
            "R01.2": {
                "status": "DONE",
                "basis": (
                    "Existing C069 registration and three inputs "
                    "consumed without reconstruction"
                )
            },
            "R01.3": {
                "status": "DONE" if r013_complete else "OPEN",
                "basis": (
                    "Targeted annotated binding, scope and "
                    "duplicate checks recorded with explicit limits"
                )
            },
            "R01.4": {
                "status": "DONE_FOR_THIS_PACKAGE",
                "basis": (
                    "Explicit paths and hashes used; "
                    "unresolved checks listed"
                )
            }
        },
        "roadmap_action_closed": r013_complete,
        "checklist_file_edited": False,
        "previous_results_rebuilt": False,
        "new_api_calls": 0,
        "new_model_pairs": 0,
        "new_semantic_judgments": 0,
        "metrics_calculated": False,
        "historical_gold_modified": False,
        "gate_changed": False,
        "next_step": (
            "R02_CONSUME_ACTUAL_ASTRA_REVIEW_AND_RECORD_METHOD_DECISIONS"
            if r013_complete else
            "RESOLVE_ONLY_REPORTED_R01_3_BINDING_OR_REPORT_DIFFERENCES"
        )
    }

    payload = (
        json.dumps(
            result,
            ensure_ascii=False,
            sort_keys=True,
            indent=2,
            allow_nan=False
        ) + "\n"
    ).encode("utf-8")
    digest = sha(payload)

    out_dir.mkdir(parents=True, exist_ok=True)
    target = out_dir / ("C002.R01." + digest + ".json")

    if target.exists():
        need(
            target.read_bytes() == payload,
            "OUTPUT_CONFLICT: " + str(target)
        )
        write_status = "REUSED_IDENTICAL"
    else:
        with target.open("xb") as handle:
            handle.write(payload)
        write_status = "WRITTEN"

    disk = target.read_bytes()
    need(
        sha(disk) == digest and parse(disk) == result,
        "RELOAD_FAILED: " + str(target)
    )

    # Actual keys and pointers only; no raw passage/request text printed.
    examples = []
    for row in annotation_results:
        if (
            row["status"]
            == "EXACT_SOURCE_AND_PROPOSAL_CONTENT_BOUND"
        ):
            continue

        eid = row["evidence_id"]
        examples.append({
            **row,
            "observed_nodes": [
                {
                    "pointer": pointer,
                    "keys": sorted(node)
                }
                for pointer, node
                in annotated_nodes.get(eid, [])[:4]
            ]
        })
        if len(examples) == 2:
            break

    print(json.dumps({
        "status": result["status"],
        "roadmap_action": result["roadmap_action"],
        "snapshot_scope": result["snapshot_scope"],
        "queries_accounted": result["queries_accounted"],
        "pairs_accounted": result["pairs_accounted"],
        "C070_C112_snapshot_comparison": snapshot_comparison,
        "subtasks": result["subtasks"],
        "checks": result["checks"],
        "pending": pending,
        "unresolved_annotated_examples": examples,
        "roadmap_action_closed": r013_complete,
        "new_api_calls": 0,
        "new_model_pairs": 0,
        "new_semantic_judgments": 0,
        "metrics_calculated": False,
        "next_step": result["next_step"],
        "write_status": write_status,
        "output": {
            "path": str(target.relative_to(root)),
            "file_bytes_sha256": digest,
            "reload_validation": "PASS"
        }
    }, ensure_ascii=False, indent=2))

    return result


V3_C002 = run_v3_c002()
````

````python
# NB22E-v3-C006 — Consume existing R04 evidence without generating new verdicts.
# Uses C001 imports. All historical files remain unchanged.

def run_v3_c006():
    required = ("os", "json", "hashlib", "Path", "Counter")
    missing = [name for name in required if name not in globals()]
    if missing:
        raise RuntimeError(
            "ابتدا C001 را اجرا کنید. نام‌های تعریف‌نشده: "
            + ", ".join(missing)
        )

    root = Path(os.environ.get(
        "RAG_FINANCE_ROOT",
        "/Users/macbookpro/AMIR_DATA/00_WORKS/01_PROJECTS/"
        "RAG-projects/RAG_finance"
    )).expanduser().resolve()
    out_dir = root / "reports/phase2/quality_improvement/NB22E.lab.v3"

    def need(condition, message):
        if not condition:
            raise ValueError(message)

    def unique_pairs(items):
        obj = {}
        for key, value in items:
            need(key not in obj, "DUPLICATE_JSON_KEY: " + key)
            obj[key] = value
        return obj

    def bad_constant(value):
        raise ValueError("NONFINITE_JSON: " + value)

    def parse(raw):
        return json.loads(
            raw.decode("utf-8"),
            object_pairs_hook=unique_pairs,
            parse_constant=bad_constant
        )

    def canonical(value):
        return json.dumps(
            value, ensure_ascii=False, sort_keys=True,
            separators=(",", ":"), allow_nan=False
        ).encode("utf-8")

    def sha(raw):
        return hashlib.sha256(raw).hexdigest()

    receipts = {}

    def read_pinned(label, relative_path, digest):
        path = (root / relative_path).resolve()
        need(
            path.is_relative_to(root),
            "PATH_OUTSIDE_PROJECT: " + str(path)
        )
        if not path.is_file():
            raise FileNotFoundError(
                "REQUIRED_INPUT_MISSING: " + str(path)
            )
        raw = path.read_bytes()
        need(sha(raw) == digest, "HASH_MISMATCH: " + str(path))
        value = parse(raw)
        receipts[label] = {
            "path": str(path.relative_to(root)),
            "file_bytes_sha256": digest
        }
        return value

    parent_sha = (
        "989cc5ead45f6520cc8d575d59058ffa"
        "b8e86be97fb12e50f657770c130966ec"
    )
    parent = read_pinned(
        "C005_R03",
        "reports/phase2/quality_improvement/NB22E.lab.v3/"
        "C005.R03_method_contract." + parent_sha + ".json",
        parent_sha
    )
    need(
        parent["method_preparation_complete_for_diagnostic_use"] is True
        and parent["addendum"]["consumption_mode"]
        == "DIAGNOSTIC_MODEL_ASSISTED_PROVISIONAL",
        "R03_DIAGNOSTIC_METHOD_NOT_READY"
    )

    latest_sha = (
        "565cd36340044a6c879045db52d6ac8d"
        "f74eea8e1e3823556202b4541a5d1abf"
    )
    ledger = read_pinned(
        "C112_existing_ledger",
        "reports/phase2/quality_improvement/NB22E.lab.v2/"
        "C112.review_work_queue." + latest_sha + ".json",
        latest_sha
    )

    pair_rows = ledger["pairs"]
    queue = ledger["targeted_review_queue"]
    need(
        isinstance(pair_rows, list) and isinstance(queue, dict),
        "PAIR_OR_QUEUE_SCHEMA_CONFLICT"
    )

    pairs = {}
    for row in pair_rows:
        eid = row["evidence_id"]
        need(eid not in pairs, "DUPLICATE_PAIR: " + eid)
        pairs[eid] = row

    need(
        len(pairs) == 197
        and len({row["case_id"] for row in pairs.values()}) == 12,
        "EXPECTED_197_PAIRS_AND_12_QUESTIONS"
    )
    need(
        len(queue) == 106 and set(queue).issubset(pairs),
        "EXISTING_QUEUE_IDENTITY_CONFLICT"
    )

    # Known prior sections read from the actual producer code.
    # Absence is recorded, never replaced with fabricated evidence.
    section_names = (
        "case01_secondary_review",
        "scope_secondary_reviews",
        "context_dependency_review",
        "case09_context_resolution_and_R05_handoff",
        "targeted_relevance_proposals_C091",
        "final_diagnostic_decision",
        "final_checklist_disposition",
        "final_coverage_disposition"
    )
    section_receipts = {}
    for name in section_names:
        if name in ledger:
            section_receipts[name] = {
                "status": "EXISTING_SECTION_CONSUMED",
                "canonical_sha256": sha(canonical(ledger[name]))
            }
        else:
            section_receipts[name] = {
                "status": "SECTION_NOT_PRESENT_IN_THIS_LEDGER"
            }

    # Bind only pair-level records from known pair-review sections.
    review_refs = {eid: [] for eid in pairs}
    review_conflicts = []

    def bind_review(eid, record, pointer, require_passage_hash=False):
        need(eid in pairs, "REVIEW_PAIR_NOT_IN_POPULATION: " + eid)
        need(isinstance(record, dict), "REVIEW_RECORD_SCHEMA: " + pointer)
        original = pairs[eid]["original_proposal"]
        if "evidence_id" in record:
            need(
                record["evidence_id"] == eid,
                "REVIEW_EVIDENCE_ID_CONFLICT: " + pointer
            )

        conflicts = []
        if "case_id" in record:
            if record["case_id"] != pairs[eid]["case_id"]:
                conflicts.append("CASE_ID_CONFLICT")
        if "passage_sha256" in record:
            if record["passage_sha256"] != original["passage_sha256"]:
                conflicts.append("PASSAGE_HASH_CONFLICT")
        elif require_passage_hash:
            conflicts.append("PASSAGE_HASH_NOT_PRESENT")

        ref = {
            "ledger_pointer": pointer,
            "record_canonical_sha256": sha(canonical(record)),
            "record": record,
            "binding_status": (
                "RECORDED_IDENTITY_AND_AVAILABLE_VERSION_FIELDS_MATCH"
                if not conflicts else "BINDING_UNRESOLVED"
            ),
            "passage_hash_explicitly_matched": (
                "passage_sha256" in record
                and record["passage_sha256"] == original["passage_sha256"]
            ),
            "binding_issues": conflicts,
            "semantic_accuracy_revalidated": False
        }
        review_refs[eid].append(ref)
        if conflicts:
            review_conflicts.append({
                "evidence_id": eid,
                "pointer": pointer,
                "issues": conflicts
            })

    for section_name in ("case01_secondary_review", "scope_secondary_reviews"):
        if section_name not in ledger:
            continue
        reviews = ledger[section_name]["reviews"]
        need(
            isinstance(reviews, dict),
            "EXPECTED_REVIEW_MAPPING: " + section_name
        )
        for eid, review in reviews.items():
            bind_review(
                eid, review, "/" + section_name + "/reviews/" + eid
            )

    if "targeted_relevance_proposals_C091" in ledger:
        seen_c091 = set()
        decisions = ledger["targeted_relevance_proposals_C091"]["decisions"]
        need(isinstance(decisions, list), "C091_DECISION_SCHEMA")
        for index, review in enumerate(decisions):
            eid = review["evidence_id"]
            need(eid not in seen_c091, "DUPLICATE_C091_PAIR: " + eid)
            seen_c091.add(eid)
            bind_review(
                eid, review,
                "/targeted_relevance_proposals_C091/decisions/" + str(index),
                require_passage_hash=True
            )

    context = ledger.get("context_dependency_review")
    context_ids = set()
    if isinstance(context, dict):
        for field in ("dependent_evidence_id", "context_evidence_id"):
            if field in context:
                eid = context[field]
                need(eid in pairs, "CONTEXT_PAIR_NOT_IN_POPULATION: " + eid)
                context_ids.add(eid)

    records = []
    counts = Counter()
    unresolved_reason_counts = Counter()

    for eid, row in sorted(pairs.items()):
        original = row["original_proposal"]
        need(
            original["evidence_id"] == eid
            and original["case_id"] == row["case_id"],
            "ORIGINAL_PROPOSAL_IDENTITY_CONFLICT: " + eid
        )
        need(
            original["approved_for_evaluation"] is False
            and original["approved_relevance_label"] is None,
            "UNEXPECTED_ORIGINAL_ACCEPTANCE_STATE: " + eid
        )
        need(
            isinstance(original.get("reason"), str)
            and original["reason"].strip(),
            "ORIGINAL_REASON_MISSING: " + eid
        )

        reasons = []
        if original["proposed_retrieval_relevance"] == "UNCERTAIN":
            reasons.append("ORIGINAL_RELEVANCE_UNCERTAIN")
        if original["proposed_answer_support"] == "UNCERTAIN":
            reasons.append("ORIGINAL_SUPPORT_UNCERTAIN")
        if original["question_scope_unresolved"]:
            reasons.append("ORIGINAL_SCOPE_FLAG_PRESERVED_NOT_A_FINAL_VERDICT")

        linked = review_refs[eid]
        labels = sorted({
            review["record"]["secondary_relevance_proposal"]
            for review in linked
            if review["binding_status"]
            == "RECORDED_IDENTITY_AND_AVAILABLE_VERSION_FIELDS_MATCH"
            and review["record"].get("secondary_relevance_proposal")
            in {"RELEVANT", "NOT_RELEVANT", "UNCERTAIN"}
        })
        if len(labels) > 1:
            reasons.append("MULTIPLE_EXISTING_SECONDARY_LABELS_REQUIRE_DISPOSITION")
        if any(review["binding_issues"] for review in linked):
            reasons.append("EXISTING_REVIEW_BINDING_UNRESOLVED")

        in_queue = eid in queue
        if linked:
            status = "EXISTING_PAIR_REVIEWS_CONSUMED_NOT_AUTOMATICALLY_ADOPTED"
        elif eid in context_ids:
            status = "EXISTING_CONTEXT_REVIEW_CONSUMED_SINGLE_PAIR_VERDICT_NOT_INFERRED"
        elif in_queue:
            status = "EXISTING_FLAGGED_PROPOSAL_PRESERVED_WITH_RECORDED_REASONS"
        else:
            status = "EXISTING_PROPOSAL_PRESERVED_OUTSIDE_TARGETED_QUEUE"

        counts[status] += 1
        unresolved_reason_counts.update(reasons)

        records.append({
            "evidence_id": eid,
            "case_id": row["case_id"],
            "query_id": row["query_id"],
            "chunk_id": row["chunk_id"],
            "passage_sha256": original["passage_sha256"],
            "original_proposal": original,
            "original_proposal_canonical_sha256": sha(canonical(original)),
            "in_existing_targeted_queue": in_queue,
            "existing_queue_reasons": queue.get(eid, []),
            "dependency_ids_as_recorded": row["dependency_ids_as_recorded"],
            "dependency_assessment_as_recorded": row["dependency_assessment"],
            "existing_pair_review_references": linked,
            "existing_secondary_labels_not_adopted": labels,
            "context_review_reference": (
                "/context_dependency_review" if eid in context_ids else None
            ),
            "accounting_status": status,
            "preserved_issues_not_new_verdicts": reasons,
            "review_provenance": "EXISTING_MODEL_ASSISTED_PROPOSALS",
            "new_blind_independent_review": False,
            "formal_acceptance_granted": False,
            "new_semantic_decision": False
        })

    need(len(records) == 197, "FINAL_PAIR_ACCOUNTING_CONFLICT")

    # Record prior scope decisions as scoped records, not as pair verdicts.
    scoped_evidence = {}
    for name in (
        "context_dependency_review",
        "case09_context_resolution_and_R05_handoff",
        "final_diagnostic_decision",
        "final_checklist_disposition",
        "final_coverage_disposition"
    ):
        if name in ledger:
            scoped_evidence[name] = ledger[name]

    subtasks = {
        "R04.1": "DONE_EXISTING_197_PROPOSALS_CONSUMED",
        "R04.2": "NOT_TRIGGERED_NO_NEW_BLIND_REVIEW",
        "R04.3": "DONE_EXISTING_MODEL_PROPOSAL_PROVENANCE_PRESERVED",
        "R04.4": (
            "EXISTING_TARGETED_WORK_CONSUMED_RESIDUAL_SEMANTIC_"
            "DISPOSITIONS_NOT_AUTOMATICALLY_COMPLETED"
        ),
        "R04.5": (
            "EXISTING_REASONS_QUOTES_DEPENDENCIES_AND_ACCEPTANCE_"
            "STATES_RECORDED_NO_NEW_ACCEPTANCE"
        ),
        "R04.6": "DONE_UNRESOLVED_ITEMS_AND_DISAGREEMENTS_PRESERVED",
        "R03.4": (
            "ALL_197_PAIRS_ACCOUNTED_FINAL_SEMANTIC_DISPOSITION_"
            "NOT_ESTABLISHED_FOR_ALL_PAIRS"
        )
    }

    result = {
        "version": "NB22E_V3_C006_R04_EXISTING_REVIEW_CONSUMPTION_V1",
        "roadmap_action": "R04",
        "status": "EXISTING_R04_EVIDENCE_CONSUMED_WITH_RESIDUALS_PRESERVED",
        "input_receipts": receipts,
        "consumption_mode": parent["addendum"]["consumption_mode"],
        "queries_accounted": 12,
        "pairs_accounted": len(records),
        "existing_queue_pairs": len(queue),
        "accounting_status_counts": dict(counts),
        "preserved_issue_counts_nonadditive": dict(unresolved_reason_counts),
        "existing_section_receipts": section_receipts,
        "review_binding_issues": review_conflicts,
        "scoped_existing_decisions": scoped_evidence,
        "pairs": records,
        "subtasks": subtasks,
        "bounded_existing_evidence_consumption_complete": True,
        "all_R04_and_R03_4_checkboxes_completed": False,
        "all_pair_semantic_dispositions_established": False,
        "formal_reference_qualified": False,
        "existing_C069_C078_package_reopened": False,
        "question_level_review_applied_as_pair_verdict": False,
        "new_api_calls": 0,
        "new_model_pairs": 0,
        "new_semantic_judgments": 0,
        "metrics_calculated": False,
        "historical_gold_modified": False,
        "original_checklist_edited": False,
        "gate_changed": False,
        "next_package": {
            "roadmap_action": "R05",
            "task": (
                "Consume existing coverage and metric-specific dispositions "
                "while carrying forward unresolved R04/R03.4 items explicitly"
            ),
            "unresolved_items_automatically_cleared": False,
            "formal_scoring_authorized": False
        }
    }

    payload = (
        json.dumps(
            result, ensure_ascii=False, sort_keys=True,
            indent=2, allow_nan=False
        ) + "\n"
    ).encode("utf-8")
    digest = sha(payload)
    out_dir.mkdir(parents=True, exist_ok=True)
    target = out_dir / ("C006.R04_existing_reviews." + digest + ".json")

    if target.exists():
        need(target.read_bytes() == payload, "OUTPUT_CONFLICT")
        write_status = "REUSED_IDENTICAL"
    else:
        with target.open("xb") as handle:
            handle.write(payload)
        write_status = "WRITTEN"

    disk = target.read_bytes()
    need(
        sha(disk) == digest and parse(disk) == result,
        "OUTPUT_RELOAD_FAILED"
    )

    print(json.dumps({
        "status": result["status"],
        "roadmap_action": "R04",
        "queries_accounted": 12,
        "pairs_accounted": len(records),
        "existing_queue_pairs": len(queue),
        "accounting_status_counts": dict(counts),
        "preserved_issue_counts_nonadditive": dict(unresolved_reason_counts),
        "existing_section_receipts": section_receipts,
        "review_binding_issues": review_conflicts,
        "subtasks": subtasks,
        "all_R04_and_R03_4_checkboxes_completed": False,
        "formal_reference_qualified": False,
        "new_api_calls": 0,
        "new_model_pairs": 0,
        "new_semantic_judgments": 0,
        "metrics_calculated": False,
        "next_package": result["next_package"],
        "write_status": write_status,
        "output": {
            "path": str(target.relative_to(root)),
            "file_bytes_sha256": digest,
            "reload_validation": "PASS"
        }
    }, ensure_ascii=False, indent=2))
    return result


V3_C006 = run_v3_c006()
````

````python
# NB22E-v3-C007 — R05: consume existing coverage and metric conditions.
# Uses C001 imports; no inference, new judgments or metric recalculation.

def run_v3_c007():
    required = ("os", "json", "hashlib", "Path", "Counter")
    missing = [name for name in required if name not in globals()]
    if missing:
        raise RuntimeError(
            "ابتدا C001 را اجرا کنید. نام‌های تعریف‌نشده: "
            + ", ".join(missing)
        )

    root = Path(os.environ.get(
        "RAG_FINANCE_ROOT",
        "/Users/macbookpro/AMIR_DATA/00_WORKS/01_PROJECTS/"
        "RAG-projects/RAG_finance"
    )).expanduser().resolve()
    out_dir = root / "reports/phase2/quality_improvement/NB22E.lab.v3"

    def need(condition, message):
        if not condition:
            raise ValueError(message)

    def unique_pairs(items):
        obj = {}
        for key, value in items:
            need(key not in obj, "DUPLICATE_JSON_KEY: " + key)
            obj[key] = value
        return obj

    def bad_constant(value):
        raise ValueError("NONFINITE_JSON: " + value)

    def parse(raw):
        return json.loads(
            raw.decode("utf-8"),
            object_pairs_hook=unique_pairs,
            parse_constant=bad_constant
        )

    def canonical(value):
        return json.dumps(
            value, ensure_ascii=False, sort_keys=True,
            separators=(",", ":"), allow_nan=False
        ).encode("utf-8")

    def sha(raw):
        return hashlib.sha256(raw).hexdigest()

    receipts = {}

    def read_pinned(label, relative_path, digest):
        path = (root / relative_path).resolve()
        need(
            path.is_relative_to(root),
            "PATH_OUTSIDE_PROJECT: " + str(path)
        )
        if not path.is_file():
            raise FileNotFoundError(
                "REQUIRED_INPUT_MISSING: " + str(path)
            )
        raw = path.read_bytes()
        need(sha(raw) == digest, "HASH_MISMATCH: " + str(path))
        receipts[label] = {
            "path": str(path.relative_to(root)),
            "file_bytes_sha256": digest
        }
        return parse(raw)

    parent_sha = (
        "5be97c9a8a5f4db7626bbdbeefd3210b"
        "7aa5bf8026bb2468f40798903a12e568"
    )
    parent = read_pinned(
        "C006_R04",
        "reports/phase2/quality_improvement/NB22E.lab.v3/"
        "C006.R04_existing_reviews." + parent_sha + ".json",
        parent_sha
    )
    need(
        parent["bounded_existing_evidence_consumption_complete"] is True,
        "R04_EVIDENCE_CONSUMPTION_NOT_COMPLETE"
    )
    need(
        parent["review_binding_issues"] == [],
        "R04_REVIEW_BINDING_ISSUES_REQUIRE_RESOLUTION"
    )

    binding = parent["input_receipts"]["C112_existing_ledger"]
    ledger = read_pinned(
        "C112_existing_ledger",
        binding["path"],
        binding["file_bytes_sha256"]
    )

    original_rows = ledger["pairs"]
    need(isinstance(original_rows, list), "PAIR_SCHEMA")
    pairs = {}
    by_case = {}

    for row in original_rows:
        eid = row["evidence_id"]
        cid = row["case_id"]
        need(eid not in pairs, "DUPLICATE_PAIR: " + eid)
        pairs[eid] = row
        by_case.setdefault(cid, []).append(row)

    need(
        len(pairs) == 197 and len(by_case) == 12,
        "EXPECTED_12_QUESTIONS_AND_197_PAIRS"
    )

    carried_pairs = {
        row["evidence_id"]: row for row in parent["pairs"]
    }
    need(
        len(carried_pairs) == len(parent["pairs"])
        and set(carried_pairs) == set(pairs),
        "C006_PAIR_COVERAGE_CONFLICT"
    )
    for eid, row in pairs.items():
        prior = carried_pairs[eid]
        need(
            prior["case_id"] == row["case_id"]
            and prior["chunk_id"] == row["chunk_id"]
            and canonical(prior["original_proposal"])
            == canonical(row["original_proposal"]),
            "C006_ORIGINAL_PAIR_CONFLICT: " + eid
        )

    comparison = ledger["diagnostic_comparison"]
    need(
        comparison["status"] == "LIMITED_DIAGNOSTIC_COMPARISON_COMPLETE"
        and comparison["query_count"] == 4
        and comparison["reference_qualified"] is False,
        "C073_COMPARISON_SCOPE_CONFLICT"
    )

    frozen_cases = {}
    for row in comparison["per_case"]:
        cid = row["case_id"]
        need(cid not in frozen_cases, "DUPLICATE_C073_CASE")
        frozen_cases[cid] = row
    need(
        set(frozen_cases)
        == {"CASE-03", "CASE-07", "CASE-10", "CASE-11"},
        "C073_CASE_SET_CONFLICT"
    )

    coverage = ledger["R05_consumption_decision_C093"]
    rank_evidence = ledger["stored_rank_coverage_C092"]
    conditions = {}
    ranks = {}

    for row in coverage["per_case"]:
        cid = row["case_id"]
        need(cid not in conditions, "DUPLICATE_C093_CASE")
        conditions[cid] = row

    for row in rank_evidence["per_case"]:
        cid = row["case_id"]
        need(cid not in ranks, "DUPLICATE_C092_CASE")
        ranks[cid] = row

    expected_five = {
        "CASE-01", "CASE-02", "CASE-04", "CASE-06", "CASE-12"
    }
    need(
        set(conditions) == set(ranks) == expected_five,
        "C092_C093_CASE_SET_CONFLICT"
    )
    need(
        not set(frozen_cases) & set(conditions),
        "EXPECTED_SCENARIO_CASE_SETS_TO_BE_DISJOINT"
    )
    need(
        set(frozen_cases) | set(conditions) <= set(by_case),
        "SCENARIO_CASE_OUTSIDE_SOURCE_POPULATION"
    )

    section_names = (
        "diagnostic_comparison",
        "stored_rank_coverage_C092",
        "R05_consumption_decision_C093",
        "five_case_diagnostic_sensitivity_C094",
        "scoped_diagnostic_decision_C095",
        "R05_coverage_and_R06_R07_reuse"
    )
    section_receipts = {
        name: {
            "status": (
                "CONSUMED"
                if name in ledger else
                "NOT_PRESENT_IN_THIS_LEDGER"
            ),
            "canonical_sha256": (
                sha(canonical(ledger[name]))
                if name in ledger else None
            )
        }
        for name in section_names
    }

    records = []
    for cid, case_pairs in sorted(by_case.items()):
        original_counts = Counter(
            row["original_proposal"]["proposed_retrieval_relevance"]
            for row in case_pairs
        )
        record = {
            "case_id": cid,
            "pairs_accounted": len(case_pairs),
            "original_relevance_proposal_counts": dict(original_counts),
            "original_scope_flags_preserved": sum(
                row["original_proposal"]["question_scope_unresolved"] is True
                for row in case_pairs
            ),
            "original_uncertain_evidence_ids": sorted(
                row["evidence_id"] for row in case_pairs
                if row["original_proposal"]["proposed_retrieval_relevance"]
                == "UNCERTAIN"
            ),
            "accepted_positive_count": 0,
            "accepted_zero_reference_established": False,
            "outside_review_set_status": "UNJUDGED",
            "outside_review_set_count": None,
            "corpus_reference_completeness_established": False,
            "formal_metric_consumption": "BLOCKED_REFERENCE_NOT_QUALIFIED",
            "original_flags_are_current_final_verdicts": False
        }

        if cid in frozen_cases:
            record.update({
                "coverage_route": "EXISTING_FROZEN_FOUR_CASE_COMPARISON",
                "diagnostic_result_available": True,
                "metric_scope": "FROZEN_C073_PROPOSAL_SCENARIO_ONLY",
                "existing_comparison_pointer":
                    "/diagnostic_comparison/per_case",
                "new_scoring_permission_granted": False
            })

        elif cid in conditions:
            view = conditions[cid]
            rank_row = ranks[cid]
            need(
                view["query_id"] == rank_row["query_id"],
                "C092_C093_QUERY_CONFLICT: " + cid
            )

            pair_view = view["pair_comparison_view"]
            view_ids = [row["evidence_id"] for row in pair_view]
            need(
                len(view_ids) == len(set(view_ids))
                and set(view_ids)
                == {row["evidence_id"] for row in case_pairs}
                and view["pairs_accounted"] == len(case_pairs),
                "C093_PAIR_VIEW_COVERAGE_CONFLICT: " + cid
            )

            for item in pair_view:
                original = pairs[item["evidence_id"]]
                need(
                    item["chunk_id"] == original["chunk_id"]
                    and item["original_proposal"]
                    == original["original_proposal"][
                        "proposed_retrieval_relevance"
                    ],
                    "C093_PAIR_VERSION_CONFLICT: " + item["evidence_id"]
                )

            record.update({
                "coverage_route": "EXISTING_FIVE_CASE_CONDITIONAL_VIEW",
                "query_id": view["query_id"],
                "comparison_view_is_accepted_reference": False,
                "proposed_positive_pairs_in_existing_view":
                    view["proposed_positive_pairs_in_comparison_view"],
                "unresolved_ids_in_existing_view": view["unresolved_ids"],
                "per_arm_metric_conditions":
                    view["per_arm_metric_conditions"],
                "shared_remaining_conditions":
                    view["shared_remaining_conditions"],
                "stored_rank_coverage": rank_row["arms"],
                "existing_sensitivity_section_present":
                    "five_case_diagnostic_sensitivity_C094" in ledger,
                "new_scoring_permission_granted": False
            })

        else:
            record.update({
                "coverage_route":
                    "OUTSIDE_THE_TWO_CONSUMED_NUMERIC_SCENARIOS",
                "diagnostic_result_available_in_these_scenarios": False,
                "interpretation":
                    "Not included in these fixed scenarios; "
                    "not declared erroneous, unanswerable or devoid of prior review",
                "new_scoring_permission_granted": False
            })

        records.append(record)

    need(
        sum(row["pairs_accounted"] for row in records) == 197,
        "FINAL_PAIR_ACCOUNTING_CONFLICT"
    )

    route_counts = dict(Counter(
        row["coverage_route"] for row in records
    ))
    formal_accounting = {
        "total_questions_in_current_review_set": 12,
        "accounted_questions": len(records),
        "formally_evaluable_questions_under_C069_proposals": 0,
        "questions_blocked_for_formal_use": len(records),
        "accepted_zero_reference_questions_established": 0,
        "unreviewed_additional_questions_of_pool24": (
            "NOT_ASSESSED_BY_THIS_PACKAGE"
        ),
        "diagnostic_scenario_denominators_must_remain_separate": True,
        "blocked_questions_treated_as_zero_scores": False
    }

    subtasks = {
        "R05.1": "DONE_FOR_DECLARED_REVIEW_SET_AND_EXISTING_SCENARIOS",
        "R05.2": "DONE_OUTSIDE_SET_PRESERVED_AS_UNJUDGED",
        "R05.3": "DONE_PROPOSAL_REFERENCE_SEPARATED_FROM_CORPUS_AND_FORMAL_REFERENCE",
        "R05.4": "CONDITIONAL_EXPANSION_NOT_ACTIVATED_NO_24_QUESTION_CLAIM",
        "R05.5": (
            "ACCOUNTING_AND_EXISTING_METRIC_CONDITIONS_RECORDED_"
            "FORMAL_REFERENCE_REMAINS_BLOCKED"
        )
    }

    result = {
        "version": "NB22E_V3_C007_R05_EXISTING_COVERAGE_CONSUMPTION_V1",
        "roadmap_action": "R05",
        "status": "R05_DECLARED_SCOPE_ACCOUNTED_FORMAL_CONSUMPTION_BLOCKED",
        "input_receipts": receipts,
        "existing_section_receipts": section_receipts,
        "per_question": records,
        "coverage_route_question_counts": route_counts,
        "formal_accounting": formal_accounting,
        "subtasks": subtasks,
        "bounded_coverage_accounting_complete": True,
        "reference_qualified": False,
        "open_R04_and_R03_4_items_carried_from_C006": parent["subtasks"],
        "carried_items_automatically_closed": False,
        "corpus_reference_completeness_established": False,
        "new_24_question_review_started": False,
        "old_metric_results_recalculated": False,
        "new_api_calls": 0,
        "new_model_pairs": 0,
        "new_semantic_judgments": 0,
        "metrics_calculated": False,
        "historical_gold_modified": False,
        "gate_changed": False,
        "original_checklist_edited": False,
        "next_package": {
            "roadmap_action": "R06",
            "task": (
                "Consume stored C073 comparison and C094 sensitivity results "
                "with separate scopes; determine which original R06 "
                "requirements are satisfied, blocked or conditional"
            ),
            "new_inference": False,
            "automatic_recalculation": False,
            "formal_metric_authorization": False
        }
    }

    payload = (
        json.dumps(
            result, ensure_ascii=False, sort_keys=True,
            indent=2, allow_nan=False
        ) + "\n"
    ).encode("utf-8")
    digest = sha(payload)
    out_dir.mkdir(parents=True, exist_ok=True)
    target = out_dir / ("C007.R05_existing_coverage." + digest + ".json")

    if target.exists():
        need(target.read_bytes() == payload, "OUTPUT_CONFLICT")
        write_status = "REUSED_IDENTICAL"
    else:
        with target.open("xb") as handle:
            handle.write(payload)
        write_status = "WRITTEN"

    disk = target.read_bytes()
    need(
        sha(disk) == digest and parse(disk) == result,
        "OUTPUT_RELOAD_FAILED"
    )

    print(json.dumps({
        "status": result["status"],
        "roadmap_action": "R05",
        "coverage_route_question_counts": route_counts,
        "formal_accounting": formal_accounting,
        "subtasks": subtasks,
        "per_question": records,
        "existing_section_receipts": section_receipts,
        "carried_items_automatically_closed": False,
        "reference_qualified": False,
        "new_api_calls": 0,
        "new_model_pairs": 0,
        "new_semantic_judgments": 0,
        "metrics_calculated": False,
        "next_package": result["next_package"],
        "write_status": write_status,
        "output": {
            "path": str(target.relative_to(root)),
            "file_bytes_sha256": digest,
            "reload_validation": "PASS"
        }
    }, ensure_ascii=False, indent=2))
    return result


V3_C007 = run_v3_c007()
````

````python
# NB22E-v3-C009 — R03.4: assemble actual texts and existing reviews.
# Uses C001 imports. Does not adjudicate, approve or advance the roadmap.

def run_v3_c009():
    required = ("os", "json", "hashlib", "Path", "Counter")
    missing = [name for name in required if name not in globals()]
    if missing:
        raise RuntimeError(
            "ابتدا C001 را اجرا کنید. نام‌های تعریف‌نشده: "
            + ", ".join(missing)
        )

    root = Path(os.environ.get(
        "RAG_FINANCE_ROOT",
        "/Users/macbookpro/AMIR_DATA/00_WORKS/01_PROJECTS/"
        "RAG-projects/RAG_finance"
    )).expanduser().resolve()
    out_dir = root / "reports/phase2/quality_improvement/NB22E.lab.v3"

    def need(condition, message):
        if not condition:
            raise ValueError(message)

    def unique_pairs(items):
        obj = {}
        for key, value in items:
            need(key not in obj, "DUPLICATE_JSON_KEY: " + key)
            obj[key] = value
        return obj

    def bad_constant(value):
        raise ValueError("NONFINITE_JSON: " + value)

    def parse(raw):
        return json.loads(
            raw.decode("utf-8"),
            object_pairs_hook=unique_pairs,
            parse_constant=bad_constant
        )

    def canonical(value):
        return json.dumps(
            value, ensure_ascii=False, sort_keys=True,
            separators=(",", ":"), allow_nan=False
        ).encode("utf-8")

    def sha(raw):
        return hashlib.sha256(raw).hexdigest()

    receipts = {}

    def read_pinned(label, relative_path, digest, jsonl=False):
        path = (root / relative_path).resolve()
        need(
            path.is_relative_to(root),
            "PATH_OUTSIDE_PROJECT: " + str(path)
        )
        if not path.is_file():
            raise FileNotFoundError(
                "REQUIRED_INPUT_MISSING: " + str(path)
            )
        raw = path.read_bytes()
        need(sha(raw) == digest, "HASH_MISMATCH: " + str(path))
        receipts[label] = {
            "path": str(path.relative_to(root)),
            "file_bytes_sha256": digest
        }
        return (
            [parse(line) for line in raw.splitlines() if line.strip()]
            if jsonl else parse(raw)
        )

    parent_sha = (
        "f65b8ba62b95b3a28b3d060e1292d88e"
        "d8ec20b77dff309867084b75d13d81b7"
    )
    parent = read_pinned(
        "C008_R03_4",
        "reports/phase2/quality_improvement/NB22E.lab.v3/"
        "C008.R03_4_dispositions." + parent_sha + ".json",
        parent_sha
    )
    need(
        parent["roadmap_action"] == "R03"
        and parent["roadmap_subtask"] == "R03.4",
        "PARENT_ACTION_CONFLICT"
    )

    previous_binding = parent["input_receipts"][
        "C006_existing_pair_evidence"
    ]
    previous = read_pinned(
        "C006_existing_pair_evidence",
        previous_binding["path"],
        previous_binding["file_bytes_sha256"]
    )

    source_sha = (
        "97ef987a570cae5f6596e4173efe6488"
        "03fbf975fc59e602499f38455f3209ad"
    )
    source = read_pinned(
        "C068_original_review_sheet",
        "reports/phase2/quality_improvement/NB22E.lab.v2/"
        "C068.blind_review_sheet." + source_sha + ".jsonl",
        source_sha,
        jsonl=True
    )

    def index_unique(rows):
        indexed = {}
        for row in rows:
            eid = row["evidence_id"]
            need(eid not in indexed, "DUPLICATE_EVIDENCE_ID: " + eid)
            indexed[eid] = row
        return indexed

    parent_rows = index_unique(parent["pairs"])
    previous_rows = index_unique(previous["pairs"])
    need(
        len(parent_rows) == 197
        and set(parent_rows) == set(previous_rows),
        "EXISTING_PAIR_POPULATION_CONFLICT"
    )

    case_ids = set()
    seen_evidence = set()
    review_cases = []

    for case in sorted(source, key=lambda item: item["case_id"]):
        cid = case["case_id"]
        need(cid not in case_ids, "DUPLICATE_CASE: " + cid)
        case_ids.add(cid)

        question = case["question"]
        need(
            isinstance(question, str) and question.strip(),
            "QUESTION_TEXT_MISSING: " + cid
        )

        case_pairs = []
        for item in case["evidence"]:
            eid = item["evidence_id"]
            need(
                eid not in seen_evidence,
                "DUPLICATE_SOURCE_EVIDENCE: " + eid
            )
            seen_evidence.add(eid)
            need(eid in parent_rows, "SOURCE_PAIR_NOT_BOUND: " + eid)

            row = parent_rows[eid]
            prior = previous_rows[eid]
            original = row["original_proposal"]
            passage = item["passage"]

            need(
                row["case_id"] == prior["case_id"] == cid
                and row["chunk_id"] == prior["chunk_id"],
                "PAIR_IDENTITY_CONFLICT: " + eid
            )
            need(
                isinstance(passage, str)
                and sha(passage.encode("utf-8"))
                == original["passage_sha256"]
                == row["passage_sha256"],
                "PASSAGE_HASH_CONFLICT: " + eid
            )
            need(
                canonical(original)
                == canonical(prior["original_proposal"]),
                "ORIGINAL_PROPOSAL_CONFLICT: " + eid
            )
            quote = original["supporting_quote"]
            need(
                isinstance(quote, str)
                and bool(quote)
                and quote in passage,
                "ORIGINAL_QUOTE_BINDING_CONFLICT: " + eid
            )

            case_pairs.append({
                "evidence_id": eid,
                "chunk_id": row["chunk_id"],
                "passage_sha256": row["passage_sha256"],
                "full_source_evidence_record": item,
                "original_proposal": original,
                "existing_review_references":
                    row["existing_review_references"],
                "existing_secondary_labels_not_adopted":
                    row["existing_secondary_labels_not_adopted"],
                "existing_context_reference":
                    row["existing_context_reference"],
                "dependency_ids_as_recorded":
                    row["dependency_ids_as_recorded"],
                "dependency_assessment_as_recorded":
                    row["dependency_assessment_as_recorded"],
                "prior_C008_accounting_status_not_final_verdict":
                    row["R03_4_row_status"],
                "new_review_decision": None,
                "formal_acceptance_granted": False
            })

        review_cases.append({
            "case_id": cid,
            "question": question,
            "full_original_case_record": case,
            "pairs": case_pairs
        })

    need(
        len(case_ids) == 12
        and len(seen_evidence) == 197
        and seen_evidence == set(parent_rows),
        "FINAL_REVIEW_MATERIAL_COVERAGE_CONFLICT"
    )

    result = {
        "version": "NB22E_V3_C009_R03_4_FULL_TEXT_REVIEW_MATERIAL_V1",
        "roadmap_action": "R03",
        "roadmap_subtask": "R03.4",
        "status": "EXISTING_FULL_TEXT_AND_REVIEWS_READY_FOR_CONTENT_REVIEW",
        "input_receipts": receipts,
        "queries_accounted": len(review_cases),
        "pairs_accounted": len(seen_evidence),
        "cases": review_cases,
        "existing_scoped_decisions":
            previous["scoped_existing_decisions"],
        "interpretation_correction": {
            "C008_176_rows_are_error_count": False,
            "C008_176_rows_are_new_API_request_count": False,
            "formal_acceptance_absence_proves_semantic_uncertainty": False,
            "diagnostic_disposition_equals_formal_reference_acceptance": False
        },
        "review_requirements": [
            "READ_QUESTION_AND_FULL_PASSAGE",
            "CONSUME_EXISTING_REASONS_AND_SECONDARY_REVIEWS",
            "KEEP_RELEVANCE_SUPPORT_AND_SOURCE_TRUTH_SEPARATE",
            "DO_NOT_INVENT_HIDDEN_QUESTION_CONTEXT",
            "RECORD_REASONED_DIAGNOSTIC_DECISION_OR_SPECIFIC_UNCERTAINTY",
            "DO_NOT_AUTOMATICALLY_GRANT_FORMAL_REFERENCE_ACCEPTANCE"
        ],
        "R03_4_complete": False,
        "advance_to_next_roadmap_action": False,
        "new_api_calls": 0,
        "new_model_pairs": 0,
        "new_semantic_judgments": 0,
        "metrics_calculated": False,
        "historical_gold_modified": False,
        "next_step": "READ_THIS_SINGLE_FILE_FOR_R03_4_CONTENT_REVIEW"
    }

    payload = (
        json.dumps(
            result, ensure_ascii=False, sort_keys=True,
            indent=2, allow_nan=False
        ) + "\n"
    ).encode("utf-8")
    digest = sha(payload)

    out_dir.mkdir(parents=True, exist_ok=True)
    target = out_dir / (
        "C009.R03_4_full_text_review." + digest + ".json"
    )
    if target.exists():
        need(target.read_bytes() == payload, "OUTPUT_CONFLICT")
        write_status = "REUSED_IDENTICAL"
    else:
        with target.open("xb") as handle:
            handle.write(payload)
        write_status = "WRITTEN"

    disk = target.read_bytes()
    need(
        sha(disk) == digest and parse(disk) == result,
        "OUTPUT_RELOAD_FAILED"
    )

    print(json.dumps({
        "status": result["status"],
        "roadmap_action": "R03",
        "roadmap_subtask": "R03.4",
        "queries_accounted": len(review_cases),
        "pairs_accounted": len(seen_evidence),
        "per_case_pair_counts": {
            case["case_id"]: len(case["pairs"])
            for case in review_cases
        },
        "R03_4_complete": False,
        "advance_to_next_roadmap_action": False,
        "new_api_calls": 0,
        "new_semantic_judgments": 0,
        "metrics_calculated": False,
        "next_step": result["next_step"],
        "write_status": write_status,
        "output": {
            "path": str(target.relative_to(root)),
            "file_bytes_sha256": digest,
            "size_bytes": len(payload),
            "reload_validation": "PASS"
        },
        "user_action_fa":
            "فقط فایل JSON معرفی‌شده در output را در همین چت پیوست کنید؛ "
            "برای بررسی محتوا، خلاصهٔ چاپ‌شده جای فایل کامل را نمی‌گیرد."
    }, ensure_ascii=False, indent=2))
    return result


V3_C009 = run_v3_c009()
````

````python
# C010 — ثبت نتیجهٔ محتوایی R03.4؛ بدون import و بدون API

def run_v3_c010():
    required = ("Path", "os", "json", "hashlib")
    missing = [name for name in required if name not in globals()]
    if missing:
        raise RuntimeError(
            "سل اول imports را اجرا کنید. موارد موجودنیست: "
            + ", ".join(missing)
        )

    def need(condition, message):
        if not condition:
            raise ValueError(message)

    def canonical(value):
        return json.dumps(
            value,
            ensure_ascii=False,
            sort_keys=True,
            separators=(",", ":"),
            allow_nan=False
        ).encode("utf-8")

    def sha(data):
        return hashlib.sha256(data).hexdigest()

    def unique_object(items):
        result = {}
        for key, value in items:
            need(key not in result, "DUPLICATE_JSON_KEY: " + key)
            result[key] = value
        return result

    def reject_constant(value):
        raise ValueError("INVALID_JSON_CONSTANT: " + value)

    def read_pinned(path, expected_sha):
        need(path.is_file(), "FILE_NOT_FOUND: " + str(path))
        data = path.read_bytes()
        need(
            sha(data) == expected_sha,
            "FILE_HASH_CONFLICT: " + str(path)
        )
        value = json.loads(
            data.decode("utf-8"),
            object_pairs_hook=unique_object,
            parse_constant=reject_constant
        )
        return value, data

    def walk_objects(value, pointer=""):
        if isinstance(value, dict):
            yield pointer, value
            for key, child in value.items():
                escaped = str(key).replace("~", "~0").replace("/", "~1")
                yield from walk_objects(child, pointer + "/" + escaped)
        elif isinstance(value, list):
            for index, child in enumerate(value):
                yield from walk_objects(child, pointer + "/" + str(index))

    def find_unique_content(matches, description):
        need(bool(matches), description + "_NOT_FOUND")
        groups = {}
        for pointer, value in matches:
            fingerprint = sha(canonical(value))
            if fingerprint not in groups:
                groups[fingerprint] = {
                    "value": value,
                    "pointers": []
                }
            groups[fingerprint]["pointers"].append(pointer)

        need(
            len(groups) == 1,
            description + "_AMBIGUOUS: "
            + json.dumps(
                [pointer for pointer, _ in matches],
                ensure_ascii=False
            )
        )
        return next(iter(groups.values()))

    root = Path(os.environ.get(
        "RAG_FINANCE_ROOT",
        "/Users/macbookpro/AMIR_DATA/00_WORKS/01_PROJECTS/"
        "RAG-projects/RAG_finance"
    )).expanduser().resolve()

    out_dir = root / "reports/phase2/quality_improvement/NB22E.lab.v3"
    need(out_dir.is_dir(), "V3_DIRECTORY_NOT_FOUND: " + str(out_dir))

    source_sha = (
        "882055b5bfa822618c88f7861841d7a2b"
        "22c66c45871667f4bf876ae937264ab"
    )
    review_sha = (
        "9e575a2363983c75ffda6c29a995ba5df"
        "a5a2a1422cd80db56c17376ffc2099e"
    )
    method_sha = (
        "989cc5ead45f6520cc8d575d59058ffab"
        "8e86be97fb12e50f657770c130966ec"
    )

    source_path = out_dir / (
        "C009.R03_4_full_text_review." + source_sha + ".json"
    )
    method_path = out_dir / (
        "C005.R03_method_contract." + method_sha + ".json"
    )
    review_name = "R03_4_content_review." + review_sha + ".json"

    candidates = [
        out_dir / review_name,
        Path.home() / "Downloads" / review_name
    ]
    available = [path for path in candidates if path.is_file()]
    need(
        bool(available),
        "فایل نتیجهٔ بررسی را با همان نام کامل در Downloads قرار دهید: "
        + review_name
    )

    source, _ = read_pinned(source_path, source_sha)
    method_document, _ = read_pinned(method_path, method_sha)

    review = None
    review_bytes = None
    for candidate in available:
        checked, checked_bytes = read_pinned(candidate, review_sha)
        if review is None:
            review = checked
            review_bytes = checked_bytes

    need(isinstance(source, dict), "SOURCE_NOT_OBJECT")
    need(isinstance(method_document, dict), "METHOD_NOT_OBJECT")
    need(isinstance(review, dict), "REVIEW_NOT_OBJECT")

    need(
        review.get("source_file_bytes_sha256") == source_sha,
        "REVIEW_SOURCE_CONFLICT"
    )
    need(
        review.get("review_authority")
        == "ASSISTANT_DIAGNOSTIC_REVIEW_NOT_HUMAN_GOLD",
        "REVIEW_AUTHORITY_CONFLICT"
    )
    need(
        review.get("labels_applied_to_historical_qrels") is False,
        "HISTORICAL_QRELS_CHANGE_NOT_ALLOWED"
    )

    # فایل C005 ممکن است قرارداد و خلاصه را در پوشش‌های تو‌در‌تو نگه دارد.
    # هیچ مقدار پیش‌فرضی جای اطلاعات موجود قرار نمی‌گیرد.
    method_objects = list(walk_objects(method_document))

    mode_matches = [
        (pointer + "/consumption_mode", obj["consumption_mode"])
        for pointer, obj in method_objects
        if "consumption_mode" in obj
    ]
    mode_binding = find_unique_content(
        mode_matches, "METHOD_CONSUMPTION_MODE"
    )
    need(
        mode_binding["value"] == "DIAGNOSTIC_MODEL_ASSISTED_PROVISIONAL",
        "METHOD_SCOPE_CONFLICT"
    )

    required_method_items = ("R03.1", "R03.2", "R03.3", "R03.5")
    subtask_matches = []
    for pointer, obj in method_objects:
        candidate = obj.get("subtasks")
        if (
            isinstance(candidate, dict)
            and all(key in candidate for key in required_method_items)
        ):
            subtask_matches.append((pointer + "/subtasks", candidate))

    subtask_binding = find_unique_content(
        subtask_matches, "METHOD_SUBTASKS"
    )
    method_subtasks = subtask_binding["value"]

    # تمام کنترل‌های قرارداد پیش از نوشتن خروجی انجام می‌شوند.
    for key in required_method_items:
        previous = method_subtasks[key]
        need(
            isinstance(previous, dict),
            "METHOD_SUBTASK_NOT_OBJECT: " + key
        )
        need(
            isinstance(previous.get("status"), str)
            and previous["status"].startswith("DONE"),
            "PREVIOUS_METHOD_ITEM_NOT_DONE: " + key
        )
        need(
            "basis" in previous,
            "METHOD_SUBTASK_BASIS_MISSING: " + key
        )

    cases = source.get("cases")
    rows = review.get("rows")
    need(isinstance(cases, list), "SOURCE_CASES_NOT_LIST")
    need(isinstance(rows, list), "REVIEW_ROWS_NOT_LIST")

    original = {}
    for case in cases:
        for pair in case["pairs"]:
            evidence_id = pair["evidence_id"]
            need(
                evidence_id not in original,
                "DUPLICATE_SOURCE_ID: " + evidence_id
            )
            original[evidence_id] = (case, pair)

    need(len(cases) == 12, "QUESTION_COUNT_CONFLICT")
    need(len(original) == len(rows) == 197, "PAIR_COUNT_CONFLICT")
    need(
        len({row["evidence_id"] for row in rows}) == 197,
        "DUPLICATE_REVIEW_ID"
    )
    need(
        {row["evidence_id"] for row in rows} == set(original),
        "REVIEW_ID_SET_CONFLICT"
    )

    relevance_counts = {}
    support_counts = {}
    changed_relevance = 0
    changed_support = 0

    for row in rows:
        evidence_id = row["evidence_id"]
        case, pair = original[evidence_id]
        passage = pair["full_source_evidence_record"]["passage"]
        proposal = pair["original_proposal"]

        need(row["case_id"] == case["case_id"], evidence_id + ": CASE")
        need(row["question"] == case["question"], evidence_id + ": QUESTION")
        need(row["chunk_id"] == pair["chunk_id"], evidence_id + ": CHUNK")
        need(
            row["question_utf8_sha256"]
            == sha(case["question"].encode("utf-8")),
            evidence_id + ": QUESTION_HASH"
        )
        need(
            row["passage_sha256"] == pair["passage_sha256"]
            == sha(passage.encode("utf-8")),
            evidence_id + ": PASSAGE_HASH"
        )

        start = row["quote_start_codepoints"]
        end = row["quote_end_codepoints"]
        need(
            type(start) is int
            and type(end) is int
            and 0 <= start < end <= len(passage),
            evidence_id + ": QUOTE_OFFSETS"
        )
        need(
            passage[start:end] == row["quote"],
            evidence_id + ": QUOTE_CONTENT"
        )
        need(
            isinstance(row["reason_fa"], str)
            and bool(row["reason_fa"].strip()),
            evidence_id + ": REASON_MISSING"
        )

        refs = pair["existing_review_references"]
        need(
            row["prior_review_count"] == len(refs)
            and row["prior_reviews_canonical_sha256"] == sha(canonical(refs)),
            evidence_id + ": PRIOR_REVIEW_BINDING"
        )
        need(
            row["dependency_ids_as_recorded"]
            == pair["dependency_ids_as_recorded"],
            evidence_id + ": DEPENDENCY_BINDING"
        )
        need(
            row["original_relevance"]
            == proposal["proposed_retrieval_relevance"]
            and row["original_support"]
            == proposal["proposed_answer_support"],
            evidence_id + ": ORIGINAL_PROPOSAL_BINDING"
        )
        need(
            row["formal_approved_for_evaluation"] is False
            and row["independent_blind_review"] is False,
            evidence_id + ": ACCEPTANCE_OR_INDEPENDENCE_CONFLICT"
        )

        relevance = row["diagnostic_relevance"]
        support = row["diagnostic_answer_support"]
        need(
            relevance in {"RELEVANT", "NOT_RELEVANT", "UNCERTAIN"},
            evidence_id + ": RELEVANCE_ENUM"
        )
        need(
            support in {
                "SUFFICIENT", "PARTIAL", "CONTEXT_ONLY", "NONE", "UNCERTAIN"
            },
            evidence_id + ": SUPPORT_ENUM"
        )

        expected_disposition = (
            "REASONED_UNDETERMINED"
            if relevance == "UNCERTAIN"
            else "DETERMINATE_IN_DIAGNOSTIC_SCOPE"
        )
        need(
            row["diagnostic_disposition"] == expected_disposition,
            evidence_id + ": DISPOSITION"
        )

        relevance_counts[relevance] = relevance_counts.get(relevance, 0) + 1
        support_counts[support] = support_counts.get(support, 0) + 1
        changed_relevance += relevance != row["original_relevance"]
        changed_support += support != row["original_support"]

    computed = {
        "relevance": relevance_counts,
        "support": support_counts,
        "changed_relevance_vs_original": changed_relevance,
        "changed_support_vs_original": changed_support
    }
    need(computed == review["counts"], "REVIEW_SUMMARY_CONFLICT")

    subtasks = {}
    for key in required_method_items:
        previous = method_subtasks[key]
        subtasks[key] = {
            "status": previous["status"],
            "basis": previous["basis"],
            "evidence": str(method_path.relative_to(root)),
            "source_json_pointers": [
                pointer + "/" + key
                for pointer in subtask_binding["pointers"]
            ]
        }

    subtasks["R03.4"] = {
        "status": "DONE_IN_DIAGNOSTIC_DISPOSITION_SCOPE",
        "basis_fa": (
            "تمام ۱۹۷ جفت دارای تصمیم تشخیصی یا وضعیت نامعینِ مستدل هستند؛ "
            "پذیرش رسمی و رفع همهٔ نامعین‌ها ادعا نشده است."
        ),
        "evidence": str((out_dir / review_name).relative_to(root))
    }

    uncertain_count = relevance_counts.get("UNCERTAIN", 0)

    result = {
        "status": "R03_COMPLETE_IN_DIAGNOSTIC_SCOPE",
        "roadmap_action": "R03",
        "queries_accounted": 12,
        "pairs_accounted": 197,
        "subtasks": dict(sorted(subtasks.items())),
        "counts": computed,
        "R03_4_complete_in_diagnostic_scope": True,
        "all_R03_items_complete_in_diagnostic_scope": True,
        "reasoned_uncertainties_preserved": uncertain_count,
        "formal_reference_qualified": False,
        "historical_proposals_modified": False,
        "metrics_calculated": False,
        "new_api_calls": 0,
        "new_semantic_judgments_in_this_python_cell": 0,
        "assistant_pair_reviews_registered": len(rows),
        "review_is_blind_or_human_gold": False,
        "advance_to_next_roadmap_action": False,
        "next_step": "REPORT_R03_RESULT_BEFORE_STARTING_NEXT_ACTION",
        "method_structure_binding": {
            "consumption_mode": mode_binding["value"],
            "consumption_mode_json_pointers": mode_binding["pointers"],
            "subtasks_json_pointers": subtask_binding["pointers"]
        },
        "input_receipts": {
            "C009": {
                "path": str(source_path.relative_to(root)),
                "file_bytes_sha256": source_sha
            },
            "C005": {
                "path": str(method_path.relative_to(root)),
                "file_bytes_sha256": method_sha
            },
            "content_review": {
                "path": str((out_dir / review_name).relative_to(root)),
                "file_bytes_sha256": review_sha
            }
        }
    }

    data = (
        json.dumps(
            result,
            ensure_ascii=False,
            sort_keys=True,
            indent=2,
            allow_nan=False
        ) + "\n"
    ).encode("utf-8")
    result_sha = sha(data)
    result_path = out_dir / (
        "C010.R03_content_disposition." + result_sha + ".json"
    )

    # از این نقطه به بعد، فقط ثبت فایل‌های تأییدشده انجام می‌شود.
    installed_review = out_dir / review_name
    if installed_review.exists():
        read_pinned(installed_review, review_sha)
    else:
        with installed_review.open("xb") as handle:
            handle.write(review_bytes)
        read_pinned(installed_review, review_sha)

    if result_path.exists():
        need(
            result_path.read_bytes() == data,
            "OUTPUT_CONTENT_CONFLICT"
        )
    else:
        with result_path.open("xb") as handle:
            handle.write(data)

    reloaded, _ = read_pinned(result_path, result_sha)
    need(reloaded == result, "OUTPUT_RELOAD_CONFLICT")

    print(json.dumps({
        "status": result["status"],
        "roadmap_action": "R03",
        "pairs_accounted": len(rows),
        "counts": computed,
        "all_R03_items_complete_in_diagnostic_scope": True,
        "formal_reference_qualified": False,
        "advance_to_next_roadmap_action": False,
        "method_structure_binding": result["method_structure_binding"],
        "write_status": "WRITTEN",
        "output": {
            "path": str(result_path.relative_to(root)),
            "file_bytes_sha256": result_sha,
            "reload_validation": "PASS"
        }
    }, ensure_ascii=False, indent=2))

    return result


V3_C010 = run_v3_c010()
````

````python
# C011 — R04: مصرف یکجای بررسی‌های موجود و نتیجهٔ ثبت‌شدهٔ R03
# بدون import، API، داوری تازه یا محاسبهٔ Metric

def run_v3_c011():
    missing = [
        name for name in ("Path", "os", "json", "hashlib")
        if name not in globals()
    ]
    if missing:
        raise RuntimeError(
            "ابتدا سل imports را اجرا کنید: " + ", ".join(missing)
        )

    def need(condition, message):
        if not condition:
            raise ValueError(message)

    def sha(data):
        return hashlib.sha256(data).hexdigest()

    def canonical(value):
        return json.dumps(
            value, ensure_ascii=False, sort_keys=True,
            separators=(",", ":"), allow_nan=False
        ).encode("utf-8")

    def unique_object(items):
        value = {}
        for key, item in items:
            need(key not in value, "DUPLICATE_JSON_KEY: " + key)
            value[key] = item
        return value

    def reject_constant(value):
        raise ValueError("INVALID_JSON_CONSTANT: " + value)

    def read_pinned(path, expected_sha):
        need(path.is_file(), "FILE_NOT_FOUND: " + str(path))
        data = path.read_bytes()
        need(sha(data) == expected_sha, "HASH_CONFLICT: " + str(path))
        value = json.loads(
            data.decode("utf-8"),
            object_pairs_hook=unique_object,
            parse_constant=reject_constant
        )
        need(isinstance(value, dict), "JSON_NOT_OBJECT: " + str(path))
        return value

    root = Path(os.environ.get(
        "RAG_FINANCE_ROOT",
        "/Users/macbookpro/AMIR_DATA/00_WORKS/01_PROJECTS/"
        "RAG-projects/RAG_finance"
    )).expanduser().resolve()
    directory = root / "reports/phase2/quality_improvement/NB22E.lab.v3"
    need(directory.is_dir(), "V3_DIRECTORY_NOT_FOUND")

    specifications = {
        "C006": (
            "C006.R04_existing_reviews.",
            "5be97c9a8a5f4db7626bbdbeefd3210b7aa5bf8026bb2468f40798903a12e568"
        ),
        "C009": (
            "C009.R03_4_full_text_review.",
            "882055b5bfa822618c88f7861841d7a2b22c66c45871667f4bf876ae937264ab"
        ),
        "C010": (
            "C010.R03_content_disposition.",
            "b3d2aec76970a2360844369ab8311a386d6e5ddd617213786935b8ae89b2d2b3"
        ),
        "content_review": (
            "R03_4_content_review.",
            "9e575a2363983c75ffda6c29a995ba5dfa5a2a1422cd80db56c17376ffc2099e"
        )
    }

    inputs = {}
    receipts = {}
    for name, (prefix, digest) in specifications.items():
        path = directory / (prefix + digest + ".json")
        inputs[name] = read_pinned(path, digest)
        receipts[name] = {
            "path": str(path.relative_to(root)),
            "file_bytes_sha256": digest
        }

    previous = inputs["C006"]
    source = inputs["C009"]
    registration = inputs["C010"]
    review = inputs["content_review"]

    need(
        registration.get("status") == "R03_COMPLETE_IN_DIAGNOSTIC_SCOPE"
        and registration.get("all_R03_items_complete_in_diagnostic_scope") is True,
        "R03_NOT_COMPLETE_IN_DIAGNOSTIC_SCOPE"
    )
    need(
        registration.get("formal_reference_qualified") is False,
        "UNEXPECTED_FORMAL_REFERENCE_STATE"
    )
    for name in ("C009", "content_review"):
        need(
            registration["input_receipts"][name]["file_bytes_sha256"]
            == receipts[name]["file_bytes_sha256"],
            "C010_RECEIPT_CONFLICT: " + name
        )
    need(
        review.get("source_file_bytes_sha256")
        == receipts["C009"]["file_bytes_sha256"],
        "CONTENT_REVIEW_SOURCE_CONFLICT"
    )
    need(
        review.get("review_authority")
        == "ASSISTANT_DIAGNOSTIC_REVIEW_NOT_HUMAN_GOLD"
        and review.get("labels_applied_to_historical_qrels") is False,
        "REVIEW_AUTHORITY_CONFLICT"
    )

    cases = source["cases"]
    reviewed_rows = review["rows"]
    need(len(cases) == 12, "CASE_COUNT_CONFLICT")

    originals = {}
    for case in cases:
        for pair in case["pairs"]:
            evidence_id = pair["evidence_id"]
            need(evidence_id not in originals, "DUPLICATE_SOURCE_ID")
            originals[evidence_id] = (case, pair)

    need(len(originals) == len(reviewed_rows) == 197, "PAIR_COUNT_CONFLICT")
    need(
        len({row["evidence_id"] for row in reviewed_rows}) == 197
        and {row["evidence_id"] for row in reviewed_rows} == set(originals),
        "REVIEW_ID_SET_CONFLICT"
    )

    scope_cases = {
        "CASE-02", "CASE-04", "CASE-05",
        "CASE-06", "CASE-08", "CASE-09"
    }
    need(
        scope_cases.issubset({case["case_id"] for case in cases}),
        "SCOPE_CASE_ID_CONFLICT"
    )

    ledger = []
    relevance_counts = {}
    support_counts = {}
    original_relevance_uncertain = []
    original_support_uncertain = []
    relevance_changes = []
    support_changes = []
    uncertain_ids = []

    for row in reviewed_rows:
        evidence_id = row["evidence_id"]
        case, pair = originals[evidence_id]
        proposal = pair["original_proposal"]
        passage = pair["full_source_evidence_record"]["passage"]
        refs = pair["existing_review_references"]

        need(
            row["case_id"] == case["case_id"]
            and row["question"] == case["question"]
            and row["chunk_id"] == pair["chunk_id"],
            evidence_id + ": IDENTITY_CONFLICT"
        )
        need(
            row["question_utf8_sha256"]
            == sha(case["question"].encode("utf-8"))
            and row["passage_sha256"] == pair["passage_sha256"]
            == sha(passage.encode("utf-8")),
            evidence_id + ": TEXT_HASH_CONFLICT"
        )
        need(
            row["original_relevance"]
            == proposal["proposed_retrieval_relevance"]
            and row["original_support"]
            == proposal["proposed_answer_support"],
            evidence_id + ": ORIGINAL_LABEL_CONFLICT"
        )
        need(
            row["prior_review_count"] == len(refs)
            and row["prior_reviews_canonical_sha256"] == sha(canonical(refs)),
            evidence_id + ": PRIOR_REVIEW_BINDING_CONFLICT"
        )
        need(
            row["dependency_ids_as_recorded"]
            == pair["dependency_ids_as_recorded"],
            evidence_id + ": DEPENDENCY_CONFLICT"
        )

        start = row["quote_start_codepoints"]
        end = row["quote_end_codepoints"]
        need(
            type(start) is int and type(end) is int
            and 0 <= start < end <= len(passage)
            and passage[start:end] == row["quote"],
            evidence_id + ": QUOTE_CONFLICT"
        )
        need(
            isinstance(row["reason_fa"], str)
            and bool(row["reason_fa"].strip()),
            evidence_id + ": MISSING_REASON"
        )
        need(
            row["review_mode"]
            == "SAME_ASSISTANT_CONTENT_REVIEW_AWARE_OF_PRIOR_PROPOSALS"
            and row["independent_blind_review"] is False
            and row["formal_approved_for_evaluation"] is False,
            evidence_id + ": PROVENANCE_CONFLICT"
        )

        relevance = row["diagnostic_relevance"]
        support = row["diagnostic_answer_support"]
        need(
            relevance in {"RELEVANT", "NOT_RELEVANT", "UNCERTAIN"}
            and support in {
                "SUFFICIENT", "PARTIAL", "CONTEXT_ONLY", "NONE", "UNCERTAIN"
            },
            evidence_id + ": INVALID_LABEL"
        )
        expected = (
            "REASONED_UNDETERMINED" if relevance == "UNCERTAIN"
            else "DETERMINATE_IN_DIAGNOSTIC_SCOPE"
        )
        need(
            row["diagnostic_disposition"] == expected,
            evidence_id + ": DISPOSITION_CONFLICT"
        )

        relevance_counts[relevance] = relevance_counts.get(relevance, 0) + 1
        support_counts[support] = support_counts.get(support, 0) + 1

        relevance_changed = relevance != row["original_relevance"]
        support_changed = support != row["original_support"]

        if row["original_relevance"] == "UNCERTAIN":
            original_relevance_uncertain.append(evidence_id)
        if row["original_support"] == "UNCERTAIN":
            original_support_uncertain.append(evidence_id)
        if relevance_changed:
            relevance_changes.append(evidence_id)
        if support_changed:
            support_changes.append(evidence_id)
        if relevance == "UNCERTAIN":
            uncertain_ids.append(evidence_id)

        ledger.append({
            "case_id": case["case_id"],
            "evidence_id": evidence_id,
            "chunk_id": pair["chunk_id"],
            "question": case["question"],
            "passage_sha256": pair["passage_sha256"],
            "original_proposal_preserved": proposal,
            "existing_review_references_preserved": refs,
            "existing_context_reference_preserved":
                pair.get("existing_context_reference"),
            "dependency_assessment_preserved":
                pair["dependency_assessment_as_recorded"],
            "current_diagnostic_review": row,
            "original_scope_flag_case": case["case_id"] in scope_cases,
            "relevance_changed_vs_original": relevance_changed,
            "support_changed_vs_original": support_changed,
            "current_review_supersedes_original_for_diagnostic_use_only": True,
            "historical_proposal_overwritten": False,
            "formal_approved_for_evaluation": False,
            "new_review_required_by_this_registration": False
        })

    computed = {
        "relevance": relevance_counts,
        "support": support_counts,
        "changed_relevance_vs_original": len(relevance_changes),
        "changed_support_vs_original": len(support_changes)
    }
    need(
        computed == review["counts"] == registration["counts"],
        "SUMMARY_CONFLICT"
    )
    need(
        len(original_relevance_uncertain) == 45
        and len(original_support_uncertain) == 41,
        "HISTORICAL_UNCERTAIN_COUNTS_CONFLICT"
    )

    # دفتر C006 حفظ می‌شود؛ وضعیت باز آن با نتیجهٔ بعدی اشتباه نمی‌شود.
    result = {
        "schema_version": "NB22E_V3_C011_R04_V1",
        "status": "R04_COMPLETE_IN_DIAGNOSTIC_SCOPE",
        "roadmap_action": "R04",
        "input_receipts": receipts,
        "queries_accounted": len(cases),
        "pairs_accounted": len(ledger),
        "consumption_mode": "DIAGNOSTIC_MODEL_ASSISTED_PROVISIONAL",
        "counts": computed,
        "historical_review_status_preserved": {
            "source": "C006",
            "registered_status": previous.get("status"),
            "canonical_sha256": sha(canonical(previous)),
            "earlier_open_status_is_not_current_final_status": True
        },
        "historical_attention_sets_not_new_queues": {
            "original_relevance_uncertain_ids": original_relevance_uncertain,
            "original_support_uncertain_ids": original_support_uncertain,
            "original_scope_cases": sorted(scope_cases),
            "sets_are_nonadditive": True,
            "all_members_bound_to_current_diagnostic_dispositions": True
        },
        "changes_vs_original": {
            "relevance_ids": relevance_changes,
            "support_ids": support_changes,
            "originals_preserved": True,
            "changes_are_not_independent_adjudication": True
        },
        "subtasks": {
            "R04.1": {
                "status": "DONE",
                "basis_fa": "تمام ۱۹۷ پیشنهاد اولیه با بررسی‌های متصل حفظ و مصرف شدند."
            },
            "R04.2": {
                "status": "NOT_APPLICABLE_NO_NEW_BLIND_REVIEW",
                "checkbox_completion_granted": False,
                "basis_fa": "بررسی مستقل کور تازه اجرا نشده؛ این شرط در این بسته موضوعیت ندارد."
            },
            "R04.3": {
                "status": "DONE",
                "basis_fa": "منشأ تمام بررسی‌های جاری، آگاه از پیشنهادهای قبلی ثبت و کنترل شد."
            },
            "R04.4": {
                "status": "DONE_IN_DIAGNOSTIC_SCOPE",
                "basis_fa": (
                    "نامعین‌های اولیهٔ ارتباط و پشتیبانی، شش پروندهٔ دامنه "
                    "و تغییرات رأی به تصمیم‌های محتوایی C010 متصل شدند؛ "
                    "پرچم اولیه صف تازه تلقی نشد."
                )
            },
            "R04.5": {
                "status": "DONE_IN_DIAGNOSTIC_SCOPE",
                "basis_fa": (
                    "نقل‌قول، دلیل، وابستگی ثبت‌شده، منشأ و عدم پذیرش رسمی "
                    "برای هر ۱۹۷ ردیف حفظ و کنترل شدند؛ "
                    "وابستگی تازه‌ای با کد استنتاج نشد."
                )
            },
            "R04.6": {
                "status": "DONE",
                "basis_fa": "نامعین‌های مستدل و آرای قبلی بدون تأیید مصنوعی حفظ شدند."
            }
        },
        "all_applicable_R04_items_complete_in_diagnostic_scope": True,
        "conditional_items_not_triggered": ["R04.2"],
        "reasoned_uncertainties_preserved": {
            "count": len(uncertain_ids),
            "evidence_ids": uncertain_ids,
            "automatic_negative_conversion": False,
            "automatic_new_review_queue": False
        },
        "formal_reference_qualified": False,
        "human_gold_established": False,
        "independent_blind_review_established": False,
        "locator_authenticity_newly_established": False,
        "historical_proposals_modified": False,
        "metrics_calculated": False,
        "new_api_calls": 0,
        "new_semantic_judgments": 0,
        "new_model_pairs": 0,
        "advance_to_next_roadmap_action": False,
        "next_step": "REPORT_R04_RESULT_BEFORE_STARTING_R05",
        "pair_ledger": ledger
    }

    data = (
        json.dumps(
            result, ensure_ascii=False, sort_keys=True,
            indent=2, allow_nan=False
        ) + "\n"
    ).encode("utf-8")
    digest = sha(data)
    output_path = directory / (
        "C011.R04_review_dispositions." + digest + ".json"
    )

    if output_path.exists():
        need(output_path.read_bytes() == data, "OUTPUT_CONTENT_CONFLICT")
    else:
        with output_path.open("xb") as handle:
            handle.write(data)

    reloaded = read_pinned(output_path, digest)
    need(reloaded == result, "OUTPUT_RELOAD_CONFLICT")

    print(json.dumps({
        "status": result["status"],
        "roadmap_action": "R04",
        "queries_accounted": len(cases),
        "pairs_accounted": len(ledger),
        "counts": computed,
        "subtasks": result["subtasks"],
        "all_applicable_R04_items_complete_in_diagnostic_scope": True,
        "conditional_items_not_triggered": ["R04.2"],
        "reasoned_uncertainties_preserved": len(uncertain_ids),
        "formal_reference_qualified": False,
        "new_api_calls": 0,
        "new_semantic_judgments": 0,
        "metrics_calculated": False,
        "advance_to_next_roadmap_action": False,
        "write_status": "WRITTEN",
        "output": {
            "path": str(output_path.relative_to(root)),
            "file_bytes_sha256": digest,
            "reload_validation": "PASS"
        }
    }, ensure_ascii=False, indent=2))

    return result


V3_C011 = run_v3_c011()
````

### کار پنج — تعیین پوشش مرجع و مخرج سنجه‌ها

````python
# C012 — R05: دفتر پوشش و مخرج‌های مستقل
# بدون import، API، داوری تازه یا محاسبهٔ Metric

def run_v3_c012():
    missing = [
        name for name in ("Path", "os", "json", "hashlib")
        if name not in globals()
    ]
    if missing:
        raise RuntimeError(
            "ابتدا سل imports را اجرا کنید: " + ", ".join(missing)
        )

    def need(condition, message):
        if not condition:
            raise ValueError(message)

    def sha(data):
        return hashlib.sha256(data).hexdigest()

    def canonical(value):
        return json.dumps(
            value, ensure_ascii=False, sort_keys=True,
            separators=(",", ":"), allow_nan=False
        ).encode("utf-8")

    def unique_object(items):
        result = {}
        for key, value in items:
            need(key not in result, "DUPLICATE_JSON_KEY: " + key)
            result[key] = value
        return result

    def reject_constant(value):
        raise ValueError("INVALID_JSON_CONSTANT: " + value)

    def read_pinned(path, expected_sha):
        need(path.is_file(), "FILE_NOT_FOUND: " + str(path))
        data = path.read_bytes()
        need(sha(data) == expected_sha, "HASH_CONFLICT: " + str(path))
        value = json.loads(
            data.decode("utf-8"),
            object_pairs_hook=unique_object,
            parse_constant=reject_constant
        )
        need(isinstance(value, dict), "JSON_NOT_OBJECT: " + str(path))
        return value

    root = Path(os.environ.get(
        "RAG_FINANCE_ROOT",
        "/Users/macbookpro/AMIR_DATA/00_WORKS/01_PROJECTS/"
        "RAG-projects/RAG_finance"
    )).expanduser().resolve()
    directory = root / "reports/phase2/quality_improvement/NB22E.lab.v3"
    need(directory.is_dir(), "V3_DIRECTORY_NOT_FOUND")

    specifications = {
        "C011": (
            "C011.R04_review_dispositions.",
            "76c564a1a5acb4f6aef988459e420d76665c74c05dd50a53dd9294e75a106b2a"
        ),
        "C009": (
            "C009.R03_4_full_text_review.",
            "882055b5bfa822618c88f7861841d7a2b22c66c45871667f4bf876ae937264ab"
        )
    }
    inputs = {}
    receipts = {}
    for name, (prefix, digest) in specifications.items():
        path = directory / (prefix + digest + ".json")
        inputs[name] = read_pinned(path, digest)
        receipts[name] = {
            "path": str(path.relative_to(root)),
            "file_bytes_sha256": digest
        }

    previous = inputs["C011"]
    source = inputs["C009"]

    need(
        previous.get("status") == "R04_COMPLETE_IN_DIAGNOSTIC_SCOPE"
        and previous.get(
            "all_applicable_R04_items_complete_in_diagnostic_scope"
        ) is True,
        "R04_NOT_COMPLETE"
    )
    need(
        previous["input_receipts"]["C009"]["file_bytes_sha256"]
        == receipts["C009"]["file_bytes_sha256"],
        "SOURCE_RECEIPT_CONFLICT"
    )
    need(
        previous.get("formal_reference_qualified") is False,
        "UNEXPECTED_FORMAL_REFERENCE_STATE"
    )

    cases = source["cases"]
    ledger = previous["pair_ledger"]
    need(len(cases) == 12 and len(ledger) == 197, "POPULATION_CONFLICT")

    original = {}
    case_index = {}
    for case in cases:
        case_id = case["case_id"]
        need(case_id not in case_index, "DUPLICATE_CASE_ID")
        case_index[case_id] = case
        for pair in case["pairs"]:
            evidence_id = pair["evidence_id"]
            need(evidence_id not in original, "DUPLICATE_SOURCE_PAIR")
            original[evidence_id] = (case, pair)

    need(
        len(original) == 197
        and len({row["evidence_id"] for row in ledger}) == 197
        and {row["evidence_id"] for row in ledger} == set(original),
        "PAIR_ID_SET_CONFLICT"
    )

    grouped = {case_id: [] for case_id in case_index}
    for row in ledger:
        evidence_id = row["evidence_id"]
        case, pair = original[evidence_id]
        decision = row["current_diagnostic_review"]
        need(
            row["case_id"] == case["case_id"]
            and row["chunk_id"] == pair["chunk_id"]
            and row["passage_sha256"] == pair["passage_sha256"],
            evidence_id + ": IDENTITY_CONFLICT"
        )
        need(
            row["original_proposal_preserved"] == pair["original_proposal"]
            and row["formal_approved_for_evaluation"] is False,
            evidence_id + ": PROVENANCE_CONFLICT"
        )
        need(
            decision["diagnostic_relevance"]
            in {"RELEVANT", "NOT_RELEVANT", "UNCERTAIN"},
            evidence_id + ": RELEVANCE_ENUM"
        )
        grouped[case["case_id"]].append(row)

    question_coverage = []
    for case_id in sorted(grouped):
        rows = grouped[case_id]
        labels = {
            label: [
                row["evidence_id"] for row in rows
                if row["current_diagnostic_review"]["diagnostic_relevance"]
                == label
            ]
            for label in ("RELEVANT", "NOT_RELEVANT", "UNCERTAIN")
        }
        positive = len(labels["RELEVANT"])
        uncertain = len(labels["UNCERTAIN"])

        question_coverage.append({
            "case_id": case_id,
            "question": case_index[case_id]["question"],
            "accounted_pair_count": len(rows),
            "known_positive_count": positive,
            "known_negative_count": len(labels["NOT_RELEVANT"]),
            "uncertain_count": uncertain,
            "evidence_ids_by_label": labels,
            "selected_reference_positive_denominator": {
                "status": (
                    "KNOWN_FOR_SELECTED_REVIEWED_SET_ONLY"
                    if uncertain == 0
                    else "UNRESOLVED_WITHIN_SELECTED_SET"
                ),
                "known_positive_count": positive,
                "uncertain_pair_count": uncertain,
                "exact_positive_denominator": (
                    positive if uncertain == 0 else None
                ),
                "corpus_recall_denominator_established": False
            },
            "zero_known_positive": positive == 0,
            "zero_relevant_reference_proven": positive == 0 and uncertain == 0,
            "zero_reference_scope": "SELECTED_ANNOTATED_SET_ONLY",
            "all_topk_items_judged_for_each_arm": "NOT_VERIFIED_HERE",
            "precision_at_k": {
                "evaluable": None,
                "reason_fa": (
                    "باید شناسه‌های TopK هر بازو به همین نسخهٔ تصمیم‌ها "
                    "متصل شوند؛ نامعین مؤثر در TopK نیز جدا بررسی شود."
                )
            },
            "recall_at_k": {
                "evaluable": None,
                "reason_fa": (
                    "اتصال رتبه‌ها و مخرج مثبت مرجع منتخب لازم است؛ "
                    "این مخرج، Recall کل Corpus نیست."
                )
            },
            "ndcg_at_k": {
                "evaluable": None,
                "reason_fa": (
                    "اتصال رتبه‌ها، نگاشت gain قرارداد و IDCG مرجع "
                    "لازم است؛ support خودکار gain نیست."
                )
            }
        })

    totals = {
        "questions_accounted": len(question_coverage),
        "pairs_accounted": sum(
            row["accounted_pair_count"] for row in question_coverage
        ),
        "known_positive_pairs": sum(
            row["known_positive_count"] for row in question_coverage
        ),
        "known_negative_pairs": sum(
            row["known_negative_count"] for row in question_coverage
        ),
        "uncertain_pairs": sum(
            row["uncertain_count"] for row in question_coverage
        ),
        "questions_with_uncertain_pairs": sum(
            row["uncertain_count"] > 0 for row in question_coverage
        ),
        "questions_with_zero_known_positive": sum(
            row["zero_known_positive"] for row in question_coverage
        ),
        "metric_evaluable_question_count": None,
        "metric_evaluable_count_not_yet_established": True
    }
    need(
        totals["pairs_accounted"] == 197
        and totals["known_positive_pairs"]
        == previous["counts"]["relevance"]["RELEVANT"]
        and totals["known_negative_pairs"]
        == previous["counts"]["relevance"]["NOT_RELEVANT"]
        and totals["uncertain_pairs"]
        == previous["counts"]["relevance"]["UNCERTAIN"],
        "COUNT_CONFLICT"
    )

    # حفظ زمینهٔ قبلی با دامنهٔ خودش؛ از وجود آن کفایت مرجع جدید نتیجه نمی‌شود.
    need(
        "existing_scoped_decisions" in source,
        "EXISTING_SCOPED_DECISIONS_MISSING"
    )
    scoped = source["existing_scoped_decisions"]

    result = {
        "schema_version": "NB22E_V3_C012_R05_V1",
        "status": "R05_COVERAGE_LEDGER_REGISTERED_RANK_BINDING_REMAINS",
        "roadmap_action": "R05",
        "input_receipts": receipts,
        "consumption_mode": "DIAGNOSTIC_MODEL_ASSISTED_PROVISIONAL",
        "totals": totals,
        "question_coverage": question_coverage,
        "existing_scoped_decisions_preserved": scoped,
        "existing_scoped_decisions_canonical_sha256": sha(canonical(scoped)),
        "prior_scope_decisions_automatically_extended_to_new_labels": False,
        "reference_scope": {
            "type": "SELECTED_197_PAIR_MODEL_ASSISTED_REFERENCE",
            "question_count": 12,
            "complete_corpus_reference": False,
            "complete_retrieval_pool_reference_established": False,
            "outside_selected_set": "UNJUDGED",
            "outside_set_converted_to_negative": False,
            "duplicate_chunks_merged": False,
            "automatic_expansion_to_24_questions": False,
            "formal_reference_qualified": False
        },
        "subtasks": {
            "R05.1": {
                "status": "PARTIAL_SELECTED_SET_BOUND_RANK_COVERAGE_OPEN",
                "basis_fa": (
                    "هویت تمام ۱۹۷ جفت با دفتر کار چهار تطبیق یافت؛ "
                    "پوشش مجموعهٔ لازم هر بازو و cutoff هنوز در این سل "
                    "با رتبه‌های ذخیره‌شده تطبیق داده نشده است."
                )
            },
            "R05.2": {
                "status": "DONE",
                "basis_fa": "خارج از مجموعه داوری‌نشده است؛ منفی ساخته نشد."
            },
            "R05.3": {
                "status": "DONE_SCOPE_LIMIT_DEFINED",
                "basis_fa": (
                    "دامنه فقط مرجع منتخب پیشنهادی است؛ "
                    "کفایت مرجع کل سؤال یا Corpus ادعا نشده است."
                )
            },
            "R05.4": {
                "status": "NOT_TRIGGERED_NO_24_QUESTION_EXPANSION",
                "basis_fa": "نیاز تازه‌ای به گسترش ۲۴تایی در این بسته تعریف نشد."
            },
            "R05.5": {
                "status": "PARTIAL_ACCOUNTING_DONE_METRIC_ELIGIBILITY_OPEN",
                "basis_fa": (
                    "total، مثبت شناخته‌شده، نامعین و صفرِ شناخته‌شده "
                    "جدا ثبت شدند؛ evaluable و مخرج نهایی هر سنجه "
                    "تا تطبیق رتبه و قرارداد، نامعلوم است نه صفر."
                )
            }
        },
        "pending": [{
            "code": "BIND_SAVED_RANKS_AND_METRIC_CONTRACT_TO_CURRENT_REFERENCE",
            "affected_items": ["R05.1", "R05.5"],
            "required_work_fa": (
                "از Receiptهای موجود، رتبه‌های دقیق ذخیره‌شده و قرارداد "
                "معیار را به شناسه‌های همین دفتر متصل کن؛ "
                "پوشش TopK، مثبت‌های مرجع و نامعین‌های مؤثر را "
                "برای هر سؤال و سنجه تعیین کن."
            ),
            "decision_changed_fa": (
                "تعیین می‌کند کدام سؤال برای کدام سنجه و بازو "
                "قابل‌سنجش است و مقایسهٔ مشترک چه مخرجی دارد."
            ),
            "requires_api": False,
            "requires_new_semantic_review_automatically": False,
            "requires_metric_calculation": False
        }],
        "all_R05_items_complete": False,
        "new_api_calls": 0,
        "new_semantic_judgments": 0,
        "metrics_calculated": False,
        "advance_to_next_roadmap_action": False,
        "next_step": "FINISH_R05_RANK_COVERAGE_BINDING_WITHOUT_SCORING"
    }

    data = (
        json.dumps(
            result, ensure_ascii=False, sort_keys=True,
            indent=2, allow_nan=False
        ) + "\n"
    ).encode("utf-8")
    digest = sha(data)
    output_path = directory / (
        "C012.R05_reference_coverage." + digest + ".json"
    )
    if output_path.exists():
        need(output_path.read_bytes() == data, "OUTPUT_CONTENT_CONFLICT")
    else:
        with output_path.open("xb") as handle:
            handle.write(data)

    need(read_pinned(output_path, digest) == result, "RELOAD_CONFLICT")

    print(json.dumps({
        "status": result["status"],
        "roadmap_action": "R05",
        "totals": totals,
        "subtasks": result["subtasks"],
        "pending": result["pending"],
        "all_R05_items_complete": False,
        "new_api_calls": 0,
        "new_semantic_judgments": 0,
        "metrics_calculated": False,
        "advance_to_next_roadmap_action": False,
        "write_status": "WRITTEN",
        "output": {
            "path": str(output_path.relative_to(root)),
            "file_bytes_sha256": digest,
            "reload_validation": "PASS"
        }
    }, ensure_ascii=False, indent=2))

    return result


V3_C012 = run_v3_c012()
````

### ادامهٔ کار پنج — ورودی لازم برای تطبیق پوشش رتبه‌ها

بسته شامل دفتر فعلی پوشش، کلید هویت سؤال–قطعه، قرارداد معیار
و فایل‌های موجود اتصال رتبه‌ها و مقایسه‌های قبلی است.

انتخاب خودکار «آخرین فایل» انجام نمی‌شود.
فایل‌های یک خانواده با نسخه‌های متفاوت جدا حفظ می‌شوند.
وجود فایل یا تطبیق هش، کفایت معنایی مرجع را اثبات نمی‌کند.

این سل فقط ورودی لازم برای بررسی اتصال را بسته‌بندی می‌کند.
کار پنج هنوز کامل اعلام نمی‌شود و وارد کار شش نمی‌شویم.
هیچ API، داوری تازه یا محاسبهٔ Metric انجام نمی‌شود.

پس از اجرا، فقط فایل بستهٔ معرفی‌شده در خروجی را پیوست کنید.

````python
# C013 — بستهٔ واحد ورودی‌های واقعی برای پایان تطبیق پوشش R05
# بدون import، API، inference، داوری یا محاسبهٔ Metric

def run_v3_c013():
    missing = [
        name for name in ("Path", "os", "json", "hashlib")
        if name not in globals()
    ]
    if missing:
        raise RuntimeError(
            "ابتدا سل imports را اجرا کنید: " + ", ".join(missing)
        )

    def need(condition, message):
        if not condition:
            raise ValueError(message)

    def sha(data):
        return hashlib.sha256(data).hexdigest()

    def unique_object(items):
        result = {}
        for key, value in items:
            need(key not in result, "DUPLICATE_JSON_KEY: " + key)
            result[key] = value
        return result

    def reject_constant(value):
        raise ValueError("INVALID_JSON_CONSTANT: " + value)

    def parse_json(text):
        return json.loads(
            text,
            object_pairs_hook=unique_object,
            parse_constant=reject_constant
        )

    root = Path(os.environ.get(
        "RAG_FINANCE_ROOT",
        "/Users/macbookpro/AMIR_DATA/00_WORKS/01_PROJECTS/"
        "RAG-projects/RAG_finance"
    )).expanduser().resolve()

    v2 = root / "reports/phase2/quality_improvement/NB22E.lab.v2"
    v3 = root / "reports/phase2/quality_improvement/NB22E.lab.v3"

    need(v2.is_dir(), "V2_DIRECTORY_NOT_FOUND: " + str(v2))
    need(v3.is_dir(), "V3_DIRECTORY_NOT_FOUND: " + str(v3))

    coverage_sha = (
        "259481c6bb5edf34eb70c337aa8ab303b"
        "707459248644d20a5d6469d82d52eaa"
    )
    identity_sha = (
        "6dcb4e6515f7681e18c9ae603c773078"
        "efeeb87ae7a15692bcd4ddcdfe1e9dd2"
    )
    contract_sha = (
        "edc54bb349c632e3424aceb8da881d124"
        "d681786bf0f384be0dffdfca0ae2a6b"
    )
    ledger_sha = (
        "fd4e5429ff1ffa63c0c94b980252a6dd"
        "0dfea620f326b1efa7a8641d7649d644"
    )

    exact_inputs = {
        "C012_current_coverage": (
            v3 / ("C012.R05_reference_coverage." + coverage_sha + ".json"),
            coverage_sha
        ),
        "C068_identity_key": (
            v2 / ("C068.review_identity_key." + identity_sha + ".jsonl"),
            identity_sha
        ),
        "C070_existing_ledger": (
            v2 / "C070.diagnostic_contract_and_coverage.v1.json",
            ledger_sha
        ),
        "C025_metric_contract": (
            root / "reports/phase2/evaluation/NB21C.C025.action12_contract"
            / ("C025.execution_contract." + contract_sha + ".json"),
            contract_sha
        )
    }

    files = []
    included_paths = set()

    def add_file(path, role, expected_sha=None):
        resolved = path.resolve()
        need(resolved.is_relative_to(root), "PATH_OUTSIDE_PROJECT")
        need(resolved.is_file(), "FILE_NOT_FOUND: " + str(resolved))

        relative = str(resolved.relative_to(root))
        if relative in included_paths:
            return

        raw = resolved.read_bytes()
        digest = sha(raw)
        if expected_sha is not None:
            need(
                digest == expected_sha,
                "PINNED_HASH_CONFLICT: " + relative
            )

        # الاسم المحتوي على SHA يُراجع، ولا يُعد وحده إثبات منشأ.
        embedded_hashes = [
            part.lower()
            for part in resolved.name.split(".")
            if len(part) == 64
            and all(char in "0123456789abcdefABCDEF" for char in part)
        ]
        for embedded in embedded_hashes:
            need(
                embedded == digest,
                "FILENAME_HASH_CONFLICT: " + relative
            )

        text = raw.decode("utf-8")
        if resolved.suffix == ".jsonl":
            content = []
            for number, line in enumerate(text.splitlines(), start=1):
                if line.strip():
                    try:
                        content.append(parse_json(line))
                    except (ValueError, TypeError) as error:
                        raise ValueError(
                            relative + ": JSONL_LINE_" + str(number)
                        ) from error
            content_type = "JSONL_RECORDS"
        else:
            content = parse_json(text)
            content_type = "JSON"

        files.append({
            "path": relative,
            "role": role,
            "file_bytes_sha256": digest,
            "size_bytes": len(raw),
            "hash_pinned_in_this_cell": expected_sha is not None,
            "filename_hash_checked": bool(embedded_hashes),
            "content_type": content_type,
            "content": content
        })
        included_paths.add(relative)

    for role, (path, digest) in exact_inputs.items():
        add_file(path, role, digest)

    coverage = next(
        item["content"] for item in files
        if item["role"] == "C012_current_coverage"
    )
    need(
        coverage.get("roadmap_action") == "R05"
        and coverage.get("all_R05_items_complete") is False,
        "C012_STATE_CONFLICT"
    )

    # فقط خانواده‌های مشخصِ رتبه و اتصال و مقایسهٔ قبلی، در همان پوشه.
    # تمام نسخه‌های موجود نگهداری می‌شوند؛ جدیدترین نسخه حدس زده نمی‌شود.
    families = ("C063", "C072", "C073", "C094")
    family_paths = {}
    for family in families:
        candidates = sorted(
            path for path in v2.iterdir()
            if path.is_file()
            and path.name.startswith(family + ".")
            and path.suffix in (".json", ".jsonl")
        )
        family_paths[family] = [
            str(path.relative_to(root)) for path in candidates
        ]
        for path in candidates:
            add_file(
                path,
                family + "_EXISTING_RANK_OR_COMPARISON_INPUT"
            )

    # هیچ نتیجه‌ای از نبود یک خانواده درباره نبود اجرای تاریخی صادر نمی‌شود.
    missing_families = [
        family for family, paths in family_paths.items() if not paths
    ]

    package = {
        "schema_version": "NB22E_V3_C013_R05_BINDING_INPUTS_V1",
        "status": "R05_BINDING_INPUTS_PACKAGED_FOR_CONTENT_INSPECTION",
        "roadmap_action": "R05",
        "affected_subtasks": ["R05.1", "R05.5"],
        "files": files,
        "family_paths_in_declared_directory": family_paths,
        "families_not_found_in_declared_directory": missing_families,
        "absence_means_no_prior_execution": False,
        "latest_version_selected": False,
        "rank_binding_established_by_this_cell": False,
        "previous_comparisons_recomputed": False,
        "all_R05_items_complete": False,
        "new_api_calls": 0,
        "new_semantic_judgments": 0,
        "metrics_calculated": False,
        "advance_to_next_roadmap_action": False,
        "next_step": "INSPECT_EXACT_RANK_IDENTITY_AND_CONTRACT_CONTENT",
        "user_action_fa": (
            "فقط فایل JSON معرفی‌شده در output را در همین چت پیوست کنید."
        )
    }

    data = (
        json.dumps(
            package, ensure_ascii=False, sort_keys=True,
            indent=2, allow_nan=False
        ) + "\n"
    ).encode("utf-8")
    digest = sha(data)
    output_path = v3 / (
        "C013.R05_binding_inputs." + digest + ".json"
    )

    if output_path.exists():
        need(output_path.read_bytes() == data, "OUTPUT_CONTENT_CONFLICT")
    else:
        with output_path.open("xb") as handle:
            handle.write(data)

    reloaded_bytes = output_path.read_bytes()
    need(sha(reloaded_bytes) == digest, "OUTPUT_HASH_CONFLICT")
    need(
        parse_json(reloaded_bytes.decode("utf-8")) == package,
        "OUTPUT_RELOAD_CONFLICT"
    )

    print(json.dumps({
        "status": package["status"],
        "roadmap_action": "R05",
        "affected_subtasks": package["affected_subtasks"],
        "files_included": [
            {
                "path": item["path"],
                "file_bytes_sha256": item["file_bytes_sha256"],
                "size_bytes": item["size_bytes"]
            }
            for item in files
        ],
        "families_not_found_in_declared_directory": missing_families,
        "all_R05_items_complete": False,
        "new_api_calls": 0,
        "new_semantic_judgments": 0,
        "metrics_calculated": False,
        "advance_to_next_roadmap_action": False,
        "write_status": "WRITTEN",
        "output": {
            "path": str(output_path.relative_to(root)),
            "file_bytes_sha256": digest,
            "size_bytes": len(data),
            "reload_validation": "PASS"
        },
        "user_action_fa": package["user_action_fa"]
    }, ensure_ascii=False, indent=2))

    return package


V3_C013 = run_v3_c013()
````

````python
# C014 — پایان R05: اتصال پوشش واقعی به قابلیت سنجش
# بدون import، API، داوری تازه یا محاسبهٔ امتیاز

def run_v3_c014():
    missing = [
        name for name in ("Path", "os", "json", "hashlib")
        if name not in globals()
    ]
    if missing:
        raise RuntimeError(
            "ابتدا سل imports را اجرا کنید: " + ", ".join(missing)
        )

    def need(condition, message):
        if not condition:
            raise ValueError(message)

    def sha(data):
        return hashlib.sha256(data).hexdigest()

    def unique_object(items):
        result = {}
        for key, value in items:
            need(key not in result, "DUPLICATE_JSON_KEY: " + key)
            result[key] = value
        return result

    def reject_constant(value):
        raise ValueError("INVALID_JSON_CONSTANT: " + value)

    def parse(text):
        return json.loads(
            text,
            object_pairs_hook=unique_object,
            parse_constant=reject_constant
        )

    root = Path(os.environ.get(
        "RAG_FINANCE_ROOT",
        "/Users/macbookpro/AMIR_DATA/00_WORKS/01_PROJECTS/"
        "RAG-projects/RAG_finance"
    )).expanduser().resolve()
    directory = root / "reports/phase2/quality_improvement/NB22E.lab.v3"

    package_sha = (
        "4b10aaa2eab0e1110854fe0d2374be401"
        "ca6197f1aa7c4c9a902d1ff47ad820c"
    )
    package_path = directory / (
        "C013.R05_binding_inputs." + package_sha + ".json"
    )
    need(package_path.is_file(), "C013_NOT_FOUND")
    raw = package_path.read_bytes()
    need(sha(raw) == package_sha, "C013_HASH_CONFLICT")
    package = parse(raw.decode("utf-8"))

    # ساختار فایل‌ها اکنون از بستهٔ واقعی خوانده‌شده مشخص است.
    files = package["files"]
    paths = [item["path"] for item in files]
    need(len(paths) == len(set(paths)), "DUPLICATE_PACKAGE_PATH")

    def by_role(role):
        matches = [item for item in files if item["role"] == role]
        need(len(matches) == 1, "ROLE_NOT_UNIQUE: " + role)
        return matches[0]

    def by_name(prefix):
        matches = [
            item for item in files
            if Path(item["path"]).name.startswith(prefix)
        ]
        need(len(matches) == 1, "ARTIFACT_NOT_UNIQUE: " + prefix)
        return matches[0]

    coverage_file = by_role("C012_current_coverage")
    identity_file = by_role("C068_identity_key")
    ledger_file = by_role("C070_existing_ledger")
    contract_file = by_role("C025_metric_contract")
    queries_file = by_name("C063.full_pool24_completed_queries.")
    plan_file = by_name("C063.full_pool24_plan.")
    binding_file = by_name("C072.diagnostic_rank_binding.")
    comparison_file = by_name("C073.diagnostic_comparison.")
    sensitivity_file = by_name("C094.five_case_diagnostic_sensitivity.")

    # بایت و محتوای فایل‌های محلی دوباره به بسته متصل می‌شوند.
    receipts = {}
    for item in files:
        path = (root / item["path"]).resolve()
        need(path.is_relative_to(root), "PATH_OUTSIDE_PROJECT")
        need(path.is_file(), "INPUT_NOT_FOUND: " + item["path"])
        data = path.read_bytes()
        need(
            sha(data) == item["file_bytes_sha256"],
            "INPUT_HASH_CONFLICT: " + item["path"]
        )
        text = data.decode("utf-8")
        content = (
            [parse(line) for line in text.splitlines() if line.strip()]
            if item["content_type"] == "JSONL_RECORDS"
            else parse(text)
        )
        need(
            content == item["content"],
            "PACKAGE_CONTENT_CONFLICT: " + item["path"]
        )
        receipts[item["path"]] = item["file_bytes_sha256"]

    coverage = coverage_file["content"]
    identities = identity_file["content"]
    old_ledger = ledger_file["content"]
    contract = contract_file["content"]
    queries = queries_file["content"]
    plan = plan_file["content"]
    binding = binding_file["content"]["rank_binding"]

    need(coverage["roadmap_action"] == "R05", "WRONG_ACTION")
    need(len(identities) == 12, "IDENTITY_COUNT_CONFLICT")

    metrics = ("recall_at_k", "hit_at_k", "ndcg_at_k_binary")
    for metric in metrics:
        need(
            old_ledger["contract"]["metric_rules"][metric]
            == contract["metric_policy"][metric],
            "METRIC_CONTRACT_CONFLICT: " + metric
        )
        need(
            binding_file["content"]["contract"]["metric_rules"][metric]
            == contract["metric_policy"][metric],
            "BINDING_METRIC_CONTRACT_CONFLICT: " + metric
        )

    need(
        binding["query_receipt"]["file_bytes_sha256"]
        == queries_file["file_bytes_sha256"],
        "QUERY_RECEIPT_CONFLICT"
    )
    need(
        binding["plan_receipt"]["file_bytes_sha256"]
        == plan_file["file_bytes_sha256"],
        "PLAN_RECEIPT_CONFLICT"
    )

    identity_index = {item["case_id"]: item for item in identities}
    query_index = {item["query_id"]: item for item in queries}
    old_question_index = {
        item["case_id"]: item for item in old_ledger["questions"]
    }
    bound_case_index = {
        item["case_id"]: item for item in binding["cases"]
    }
    need(len(identity_index) == 12, "DUPLICATE_CASE_ID")
    need(len(query_index) == len(queries), "DUPLICATE_QUERY_ID")

    rows = []
    for current in coverage["question_coverage"]:
        case_id = current["case_id"]
        identity = identity_index[case_id]
        query_id = identity["query_id"]
        old_question = old_question_index[case_id]

        need(
            old_question["query_id"] == query_id
            and old_question["question"] == current["question"],
            case_id + ": QUESTION_BINDING_CONFLICT"
        )
        need(query_id in query_index, case_id + ": QUERY_NOT_IN_C063")

        mapping = {
            item["chunk_id"]: item["evidence_id"]
            for item in identity["evidence_mapping"]
        }
        need(
            len(mapping) == len(identity["evidence_mapping"]),
            case_id + ": DUPLICATE_CHUNK_ID"
        )
        labels = {}
        for label, evidence_ids in current["evidence_ids_by_label"].items():
            for evidence_id in evidence_ids:
                need(evidence_id not in labels, "DUPLICATE_LABEL_ID")
                labels[evidence_id] = label

        need(
            set(mapping.values()) == set(labels),
            case_id + ": REVIEW_SET_CONFLICT"
        )

        bge_order = query_index[query_id]["bge_order"]
        need(
            isinstance(bge_order, list)
            and all(isinstance(value, str) for value in bge_order)
            and len(bge_order) == len(set(bge_order))
            and len(bge_order) >= 10,
            case_id + ": BGE_ORDER_CONFLICT"
        )
        need(
            set(bge_order) == set(plan["candidates"][query_id]),
            case_id + ": BGE_CANDIDATE_SET_CONFLICT"
        )

        arms = {
            "pure_bge_full_pool": {
                "top10_chunk_ids": bge_order[:10],
                "rank_binding_basis": "PINNED_C063_STORED_BGE_ORDER",
                "source_file_bytes_sha256": queries_file["file_bytes_sha256"]
            }
        }

        if case_id in bound_case_index:
            prior = bound_case_index[case_id]
            need(
                prior["query_id"] == query_id
                and prior["question_binding"] == "EXACT_TEXT_MATCH",
                case_id + ": PRIOR_RANK_BINDING_CONFLICT"
            )
            for arm in ("baseline", "pure_bge_full_pool"):
                prior_top = prior["arms"][arm]["top10"]
                need(
                    [entry["rank"] for entry in prior_top]
                    == list(range(1, 11)),
                    case_id + ": STORED_RANK_POSITION_CONFLICT"
                )
                for entry in prior_top:
                    need(
                        mapping.get(entry["chunk_id"]) == entry["evidence_id"],
                        case_id + ": STORED_EVIDENCE_BINDING_CONFLICT"
                    )
            need(
                [
                    entry["chunk_id"]
                    for entry in prior["arms"]["pure_bge_full_pool"]["top10"]
                ] == bge_order[:10],
                case_id + ": BGE_C072_C063_CONFLICT"
            )
            arms["baseline"] = {
                "top10_chunk_ids": [
                    entry["chunk_id"]
                    for entry in prior["arms"]["baseline"]["top10"]
                ],
                "rank_binding_basis": "EXISTING_C072_BOUND_TOP10_REUSED",
                "source_artifact_receipt":
                    prior["arms"]["baseline"]["source_artifact"],
                "large_source_artifact_reread_in_this_cell": False
            }
        else:
            arms["baseline"] = {
                "top10_chunk_ids": None,
                "rank_binding_basis": "NOT_BOUND_IN_INSPECTED_PACKAGE",
                "source_artifact_receipt": plan["source_bindings"]["baseline"],
                "missing_binding_is_missing_execution": False
            }

        positive_count = current["known_positive_count"]
        uncertain_count = current["uncertain_count"]
        reference_ready = positive_count > 0 and uncertain_count == 0

        for arm, arm_data in arms.items():
            top = arm_data["top10_chunk_ids"]
            arm_data["cutoff_coverage"] = {}
            for cutoff in (5, 10):
                selected = None if top is None else top[:cutoff]
                unjudged = (
                    None if selected is None
                    else [chunk for chunk in selected if chunk not in mapping]
                )
                uncertain_top = (
                    None if selected is None
                    else [
                        mapping[chunk] for chunk in selected
                        if chunk in mapping
                        and labels[mapping[chunk]] == "UNCERTAIN"
                    ]
                )
                blockers = []
                if selected is None:
                    blockers.append("RANK_BINDING_NOT_ESTABLISHED_IN_PACKAGE")
                elif unjudged:
                    blockers.append("UNJUDGED_TOPK_ITEMS")
                if uncertain_count:
                    blockers.append("SELECTED_REFERENCE_HAS_UNCERTAIN_LABELS")
                if positive_count == 0:
                    blockers.append("NONEMPTY_POSITIVE_REFERENCE_NOT_ESTABLISHED")

                arm_data["cutoff_coverage"][str(cutoff)] = {
                    "unjudged_topk_chunk_ids": unjudged,
                    "uncertain_topk_evidence_ids": uncertain_top,
                    "metric_eligibility": {
                        metric: {
                            "diagnostic_evaluable": not blockers,
                            "blockers": list(blockers)
                        }
                        for metric in metrics
                    },
                    "precision_at_k": "NOT_DEFINED_IN_CONSUMED_METRIC_CONTRACT"
                }

        rows.append({
            "case_id": case_id,
            "query_id": query_id,
            "question": current["question"],
            "accounted_pairs": current["accounted_pair_count"],
            "known_positive_count": positive_count,
            "uncertain_count": uncertain_count,
            "selected_reference_positive_denominator": (
                positive_count if uncertain_count == 0 else None
            ),
            "binary_idcg_basis_positive_count": (
                positive_count if reference_ready else None
            ),
            "gain_basis": "BINARY_RETRIEVAL_RELEVANCE_NOT_ANSWER_SUPPORT",
            "reference_scope": "SELECTED_REVIEWED_SET_ONLY",
            "corpus_reference_completeness_established": False,
            "formal_evaluable": False,
            "arms": arms
        })

    eligible = {}
    common = {}
    for cutoff in (5, 10):
        eligible[str(cutoff)] = {}
        for arm in ("baseline", "pure_bge_full_pool"):
            eligible[str(cutoff)][arm] = [
                row["case_id"] for row in rows
                if row["arms"][arm]["cutoff_coverage"][str(cutoff)]
                ["metric_eligibility"]["recall_at_k"]["diagnostic_evaluable"]
            ]
        common[str(cutoff)] = sorted(
            set(eligible[str(cutoff)]["baseline"])
            & set(eligible[str(cutoff)]["pure_bge_full_pool"])
        )

    need(
        len(rows) == 12
        and sum(row["accounted_pairs"] for row in rows) == 197,
        "FINAL_POPULATION_CONFLICT"
    )
    need(
        len(eligible["10"]["baseline"]) == 4
        and len(eligible["10"]["pure_bge_full_pool"]) == 7
        and len(common["10"]) == 4,
        "INSPECTED_COVERAGE_RESULT_CONFLICT"
    )

    result = {
        "schema_version": "NB22E_V3_C014_R05_V1",
        "status": "R05_COMPLETE_WITH_EXPLICIT_DIAGNOSTIC_COVERAGE_LIMITS",
        "roadmap_action": "R05",
        "input_package_sha256": package_sha,
        "input_receipts": receipts,
        "queries_accounted": len(rows),
        "pairs_accounted": 197,
        "question_metric_coverage": rows,
        "diagnostic_eligible_case_ids_by_cutoff_and_arm": eligible,
        "paired_diagnostic_eligible_case_ids_by_cutoff": common,
        "eligibility_applies_to": list(metrics),
        "formal_reference_qualified": False,
        "scope": "SELECTED_197_PAIRS_12_QUESTIONS_NOT_CORPUS_RECALL",
        "outside_selected_set": "UNJUDGED",
        "zero_known_positive_means_no_answer": False,
        "subtasks": {
            "R05.1": {
                "status": "DONE_WITH_EXPLICIT_RANK_BINDING_LIMITS",
                "basis_fa": (
                    "مجموعهٔ لازم Top5/Top10 با مرجع منتخب تطبیق یافت؛ "
                    "اتصال BGE برای ۱۲ سؤال و baseline موجود برای چهار سؤال "
                    "ثبت شد؛ هشت اتصال احرازنشده جدا ماند."
                )
            },
            "R05.2": {"status": "DONE_OUTSIDE_SET_REMAINS_UNJUDGED"},
            "R05.3": {"status": "DONE_SELECTED_REFERENCE_SCOPE_EXPLICIT"},
            "R05.4": {"status": "NOT_TRIGGERED_NO_24_QUESTION_EXPANSION"},
            "R05.5": {
                "status": "DONE_ELIGIBLE_UNRESOLVED_AND_DENOMINATORS_SEPARATED",
                "basis_fa": (
                    "قابلیت سنجش برای هر سؤال، بازو و cutoff جدا ثبت شد؛ "
                    "مخرج نامعین null ماند و نامعین‌ها منفی نشدند."
                )
            }
        },
        "all_applicable_R05_items_complete_in_declared_scope": True,
        "all_questions_evaluable": False,
        "conditional_followup": {
            "needed_only_if_expanding_paired_comparison": True,
            "task_fa": (
                "اتصال baseline هشت سؤال دیگر از رسید دقیق موجود؛ "
                "این کار شرط مصرف مقایسهٔ بسته‌شدهٔ چهار سؤال نیست."
            ),
            "baseline_receipt": plan["source_bindings"]["baseline"],
            "uncertain_labels_automatically_queued_for_review": False
        },
        "previous_comparisons": {
            "C073_file_bytes_sha256": comparison_file["file_bytes_sha256"],
            "C094_file_bytes_sha256": sensitivity_file["file_bytes_sha256"],
            "preserved_in_C013": True,
            "recomputed": False,
            "automatically_reinterpreted_under_current_labels": False
        },
        "new_api_calls": 0,
        "new_semantic_judgments": 0,
        "metrics_calculated": False,
        "thresholds_changed": False,
        "advance_to_next_roadmap_action": False,
        "next_step": "REPORT_R05_RESULT_BEFORE_R06"
    }

    data = (
        json.dumps(
            result, ensure_ascii=False, sort_keys=True,
            indent=2, allow_nan=False
        ) + "\n"
    ).encode("utf-8")
    digest = sha(data)
    output_path = directory / (
        "C014.R05_metric_eligibility." + digest + ".json"
    )
    if output_path.exists():
        need(output_path.read_bytes() == data, "OUTPUT_CONTENT_CONFLICT")
    else:
        with output_path.open("xb") as handle:
            handle.write(data)

    saved = output_path.read_bytes()
    need(sha(saved) == digest, "OUTPUT_HASH_CONFLICT")
    need(parse(saved.decode("utf-8")) == result, "OUTPUT_RELOAD_CONFLICT")

    print(json.dumps({
        "status": result["status"],
        "roadmap_action": "R05",
        "queries_accounted": len(rows),
        "pairs_accounted": 197,
        "diagnostic_eligible_case_ids_by_cutoff_and_arm": eligible,
        "paired_diagnostic_eligible_case_ids_by_cutoff": common,
        "all_applicable_R05_items_complete_in_declared_scope": True,
        "all_questions_evaluable": False,
        "formal_reference_qualified": False,
        "new_api_calls": 0,
        "new_semantic_judgments": 0,
        "metrics_calculated": False,
        "advance_to_next_roadmap_action": False,
        "write_status": "WRITTEN",
        "output": {
            "path": str(output_path.relative_to(root)),
            "file_bytes_sha256": digest,
            "reload_validation": "PASS"
        }
    }, ensure_ascii=False, indent=2))
    return result


V3_C014 = run_v3_c014()
````

````python
# C015 — R06: مصرف مقایسه‌های موجود، کنترل سازگاری و ثبت شکاف واقعی
# بدون import، API، inference، داوری یا محاسبهٔ مجدد Metric

def run_v3_c015():
    missing = [
        name for name in ("Path", "os", "json", "hashlib")
        if name not in globals()
    ]
    if missing:
        raise RuntimeError(
            "ابتدا سل imports را اجرا کنید: " + ", ".join(missing)
        )

    def need(condition, message):
        if not condition:
            raise ValueError(message)

    def sha(data):
        return hashlib.sha256(data).hexdigest()

    def unique_object(items):
        result = {}
        for key, value in items:
            need(key not in result, "DUPLICATE_JSON_KEY: " + key)
            result[key] = value
        return result

    def reject_constant(value):
        raise ValueError("INVALID_JSON_CONSTANT: " + value)

    def parse(text):
        return json.loads(
            text,
            object_pairs_hook=unique_object,
            parse_constant=reject_constant
        )

    root = Path(os.environ.get(
        "RAG_FINANCE_ROOT",
        "/Users/macbookpro/AMIR_DATA/00_WORKS/01_PROJECTS/"
        "RAG-projects/RAG_finance"
    )).expanduser().resolve()
    directory = root / "reports/phase2/quality_improvement/NB22E.lab.v3"

    specifications = {
        "C013": (
            "C013.R05_binding_inputs.",
            "4b10aaa2eab0e1110854fe0d2374be401ca6197f1aa7c4c9a902d1ff47ad820c"
        ),
        "C014": (
            "C014.R05_metric_eligibility.",
            "223c0bf5dfebeb35226b84cb5f6569be2cc2041a5a05df166e7c30a766f205e7"
        ),
        "current_review": (
            "R03_4_content_review.",
            "9e575a2363983c75ffda6c29a995ba5dfa5a2a1422cd80db56c17376ffc2099e"
        )
    }

    inputs = {}
    receipts = {}
    for name, (prefix, digest) in specifications.items():
        path = directory / (prefix + digest + ".json")
        need(path.is_file(), "FILE_NOT_FOUND: " + str(path))
        raw = path.read_bytes()
        need(sha(raw) == digest, "HASH_CONFLICT: " + name)
        inputs[name] = parse(raw.decode("utf-8"))
        receipts[name] = {
            "path": str(path.relative_to(root)),
            "file_bytes_sha256": digest
        }

    package = inputs["C013"]
    coverage = inputs["C014"]
    current_review = inputs["current_review"]

    need(
        coverage.get("all_applicable_R05_items_complete_in_declared_scope")
        is True,
        "R05_NOT_COMPLETE_IN_DECLARED_SCOPE"
    )
    need(
        coverage["input_package_sha256"]
        == receipts["C013"]["file_bytes_sha256"],
        "C014_C013_BINDING_CONFLICT"
    )

    def artifact(prefix):
        matches = [
            item for item in package["files"]
            if Path(item["path"]).name.startswith(prefix)
        ]
        need(len(matches) == 1, "ARTIFACT_NOT_UNIQUE: " + prefix)
        item = matches[0]
        path = (root / item["path"]).resolve()
        need(path.is_relative_to(root), "PATH_OUTSIDE_PROJECT")
        need(path.is_file(), "FILE_NOT_FOUND: " + item["path"])
        raw = path.read_bytes()
        need(
            sha(raw) == item["file_bytes_sha256"],
            "ARTIFACT_HASH_CONFLICT: " + item["path"]
        )
        text = raw.decode("utf-8")
        content = (
            [parse(line) for line in text.splitlines() if line.strip()]
            if item["content_type"] == "JSONL_RECORDS"
            else parse(text)
        )
        need(content == item["content"], "PACKAGE_CONTENT_CONFLICT")
        receipts[prefix] = {
            "path": item["path"],
            "file_bytes_sha256": item["file_bytes_sha256"]
        }
        return content

    previous = artifact("C073.diagnostic_comparison.")
    sensitivity_document = artifact("C094.five_case_diagnostic_sensitivity.")
    stored_queries = artifact("C063.full_pool24_completed_queries.")

    comparison = previous["diagnostic_comparison"]
    scenario = previous["diagnostic_scenario"]
    sensitivity = sensitivity_document["five_case_diagnostic_sensitivity_C094"]

    current_rows = current_review["rows"]
    current_index = {
        row["evidence_id"]: row for row in current_rows
    }
    need(len(current_rows) == len(current_index) == 197, "REVIEW_ID_CONFLICT")

    common = coverage[
        "paired_diagnostic_eligible_case_ids_by_cutoff"
    ]["10"]
    compared_cases = [
        row["case_id"] for row in comparison["per_case"]
    ]
    need(
        len(compared_cases) == len(set(compared_cases)) == 4
        and set(compared_cases) == set(common),
        "COMPARISON_CASE_SET_CONFLICT"
    )
    need(
        comparison["cutoff"] == 10
        and comparison["query_count"] == 4
        and comparison["reference_qualified"] is False,
        "COMPARISON_SCOPE_CONFLICT"
    )

    # کنترل تمام برچسب‌های مرجع چهار سؤال؛ نه فقط برچسب‌های Top10.
    scenario_rows = scenario["pairs"]
    scenario_ids = [row["evidence_id"] for row in scenario_rows]
    expected_ids = {
        row["evidence_id"] for row in current_rows
        if row["case_id"] in common
    }
    need(
        len(scenario_ids) == len(set(scenario_ids))
        and set(scenario_ids) == expected_ids,
        "SCENARIO_REFERENCE_SET_CONFLICT"
    )

    for old in scenario_rows:
        current = current_index[old["evidence_id"]]
        need(
            current["case_id"] == old["case_id"]
            and current["chunk_id"] == old["chunk_id"]
            and current["passage_sha256"] == old["passage_sha256"],
            old["evidence_id"] + ": IDENTITY_CONFLICT"
        )
        label = current["diagnostic_relevance"]
        need(label in ("RELEVANT", "NOT_RELEVANT"), "UNEXPECTED_UNCERTAIN")
        need(
            old["scenario_binary_relevance"]
            == (1 if label == "RELEVANT" else 0),
            old["evidence_id"] + ": REFERENCE_CHANGED_REUSE_NOT_ALLOWED"
        )

    coverage_index = {
        row["case_id"]: row
        for row in coverage["question_metric_coverage"]
    }
    query_index = {row["query_id"]: row for row in stored_queries}
    need(len(query_index) == len(stored_queries), "DUPLICATE_QUERY_ID")

    historical_stored_values = []
    for item in comparison["per_case"]:
        case_id = item["case_id"]
        bound = coverage_index[case_id]
        need(item["query_id"] == bound["query_id"], "QUERY_ID_CONFLICT")

        for arm in ("baseline", "pure_bge_full_pool"):
            top = item["top10_with_scenario_gains"][arm]
            need(
                [row["rank"] for row in top] == list(range(1, 11)),
                case_id + ": RANK_POSITION_CONFLICT"
            )
            need(
                [row["chunk_id"] for row in top]
                == bound["arms"][arm]["top10_chunk_ids"],
                case_id + ": STORED_RANK_CONFLICT"
            )
            for entry in top:
                current = current_index[entry["evidence_id"]]
                need(
                    current["case_id"] == case_id
                    and current["chunk_id"] == entry["chunk_id"]
                    and entry["proposed_scenario_gain"]
                    == (
                        1 if current["diagnostic_relevance"] == "RELEVANT"
                        else 0
                    ),
                    case_id + ": GAIN_BINDING_CONFLICT"
                )

        # فقط نقل مقادیر ذخیره‌شده؛ بازتولید یا تأیید تاریخی محسوب نمی‌شود.
        stored = query_index[item["query_id"]]
        historical_stored_values.append({
            "case_id": case_id,
            "query_id": item["query_id"],
            "baseline": stored["baseline"],
            "pure_bge_full_pool": stored["pure_bge_full_pool"],
            "historical_reference_independently_bound_here": False,
            "recomputed": False
        })

    need(
        comparison["historical_reviewed_four_cell_comparison_performed"]
        is False,
        "UNEXPECTED_PRIOR_FOUR_CELL_STATE"
    )
    need(
        previous["rank_binding"][
            "historical_gold_mapping_bound_for_four_cell_comparison"
        ] is False,
        "UNEXPECTED_PRIOR_HISTORICAL_BINDING_STATE"
    )
    need(
        sensitivity["reference_qualified"] is False
        and sensitivity["fixed_C073_scenario_changed"] is False
        and sensitivity["unknown_labels_adopted"] is False,
        "SENSITIVITY_SCOPE_CONFLICT"
    )
    need(
        set(sensitivity["cases"]).isdisjoint(set(common)),
        "SENSITIVITY_POPULATION_OVERLAP_CONFLICT"
    )

    result = {
        "schema_version": "NB22E_V3_C015_R06_V1",
        "status": "R06_EXISTING_DIAGNOSTICS_REUSED_FOUR_CELL_BINDING_OPEN",
        "roadmap_action": "R06",
        "input_receipts": receipts,
        "reused_four_case_comparison": comparison,
        "four_case_current_relevance_labels_identical": True,
        "four_case_rank_identity_checked": True,
        "four_case_reference_pair_count": len(scenario_rows),
        "historical_stored_values_not_reproduced": historical_stored_values,
        "five_case_historical_sensitivity": sensitivity,
        "five_case_sensitivity_applied_to_current_review": False,
        "different_populations_merged": False,
        "subtasks": {
            "R06.1": {
                "status": "DONE_FOR_DECLARED_FOUR_CASE_COMPARISON",
                "basis_fa": (
                    "اتصال موجود رتبه‌های دو بازو با شناسه‌ها، "
                    "Top10 و نسخهٔ مرجع فعلی کنترل شد."
                )
            },
            "R06.2": {
                "status": "OPEN_HISTORICAL_REFERENCE_BINDING_REQUIRED",
                "basis_fa": (
                    "مقادیر تاریخی همان سؤال‌ها موجود است؛ "
                    "بازتولید آن‌ها با مرجع تاریخی متصل انجام نشده است."
                )
            },
            "R06.3": {
                "status": "OPEN_FOUR_CELL_COMPARISON_NOT_ESTABLISHED",
                "basis_fa": (
                    "دو خانهٔ مرجع تشخیصی موجود مصرف شدند؛ "
                    "چهارخانهٔ کامل از عددهای تاریخیِ صرف ساخته نشد."
                )
            },
            "R06.4": {
                "status": "DONE",
                "basis_fa": "نتایج فقط سناریوی تشخیصی پیشنهادهای مدل هستند."
            },
            "R06.5": {
                "status": "PARTIAL_MODEL_DELTA_AVAILABLE_REFERENCE_EFFECT_OPEN",
                "basis_fa": (
                    "اختلاف مدل روی مرجع تشخیصی از نتیجهٔ موجود مصرف شد؛ "
                    "اثر تغییر مرجع و اختلاف اثرها هنوز احراز نشده‌اند."
                )
            },
            "R06.6": {
                "status": "DONE_IN_SEPARATE_EXISTING_SCOPES",
                "basis_fa": (
                    "چهار سؤال قطعی و حساسیت پنج سؤال با فرض‌های تاریخی "
                    "جدا حفظ شدند؛ نامعین‌ها حذف یا صفر نشدند."
                )
            }
        },
        "pending": [{
            "code": "BIND_HISTORICAL_GOLD_TO_SAME_FOUR_QUERY_CHUNK_IDENTITIES",
            "affected_subtasks": ["R06.2", "R06.3", "R06.5"],
            "required_work_fa": (
                "از ورودی‌ها و قرارداد تاریخی موجود، نگاشت Gold به Chunk "
                "همین چهار سؤال را با شناسه و نسخه متصل کن؛ "
                "سپس کفایت شاهد برای بازتولید تاریخی و چهارخانه تعیین شود."
            ),
            "decision_changed_fa": (
                "امکان تفکیک توصیفی اثر تغییر مرجع از اختلاف دو بازو "
                "را مشخص می‌کند؛ مقایسهٔ تشخیصی موجود را باز نمی‌کند."
            ),
            "absence_of_formal_acceptance_is_not_this_blocker": True,
            "requires_api": False,
            "requires_new_semantic_review_automatically": False
        }],
        "all_R06_items_complete": False,
        "formal_reference_qualified": False,
        "causal_error_shares_established": False,
        "new_api_calls": 0,
        "new_model_pairs": 0,
        "new_semantic_judgments": 0,
        "metrics_calculated_in_this_cell": False,
        "previous_comparisons_recomputed": False,
        "advance_to_next_roadmap_action": False,
        "next_step": "RESOLVE_ONLY_HISTORICAL_REFERENCE_BINDING_FOR_R06"
    }

    data = (
        json.dumps(
            result, ensure_ascii=False, sort_keys=True,
            indent=2, allow_nan=False
        ) + "\n"
    ).encode("utf-8")
    digest = sha(data)
    output_path = directory / (
        "C015.R06_existing_comparisons." + digest + ".json"
    )
    if output_path.exists():
        need(output_path.read_bytes() == data, "OUTPUT_CONTENT_CONFLICT")
    else:
        with output_path.open("xb") as handle:
            handle.write(data)

    saved = output_path.read_bytes()
    need(sha(saved) == digest, "OUTPUT_HASH_CONFLICT")
    need(parse(saved.decode("utf-8")) == result, "OUTPUT_RELOAD_CONFLICT")

    print(json.dumps({
        "status": result["status"],
        "roadmap_action": "R06",
        "four_case_current_relevance_labels_identical": True,
        "four_case_comparison_reused": comparison["macro"],
        "four_case_model_delta_reused":
            comparison["delta_bge_minus_baseline"],
        "five_case_sensitivity_preserved_with_original_scope": True,
        "subtasks": result["subtasks"],
        "pending": result["pending"],
        "all_R06_items_complete": False,
        "new_api_calls": 0,
        "new_model_pairs": 0,
        "metrics_calculated_in_this_cell": False,
        "previous_comparisons_recomputed": False,
        "advance_to_next_roadmap_action": False,
        "write_status": "WRITTEN",
        "output": {
            "path": str(output_path.relative_to(root)),
            "file_bytes_sha256": digest,
            "reload_validation": "PASS"
        }
    }, ensure_ascii=False, indent=2))
    return result


V3_C015 = run_v3_c015()
````

````python
# C019 — R08: مصرف A07 موجود و اتصال کد Producer برای رسیدگی دقیق
# بدون import، API، resend یا اجرای کد نوت‌بوک قبلی

def run_v3_c019():
    missing = [
        name for name in ("Path", "os", "json", "hashlib")
        if name not in globals()
    ]
    if missing:
        raise RuntimeError(
            "ابتدا سل imports را اجرا کنید: " + ", ".join(missing)
        )

    def need(condition, message):
        if not condition:
            raise ValueError(message)

    def sha(data):
        return hashlib.sha256(data).hexdigest()

    def unique_object(items):
        result = {}
        for key, value in items:
            need(key not in result, "DUPLICATE_JSON_KEY: " + key)
            result[key] = value
        return result

    def reject_constant(value):
        raise ValueError("INVALID_JSON_CONSTANT: " + value)

    def parse(text):
        return json.loads(
            text,
            object_pairs_hook=unique_object,
            parse_constant=reject_constant
        )

    root = Path(os.environ.get(
        "RAG_FINANCE_ROOT",
        "/Users/macbookpro/AMIR_DATA/00_WORKS/01_PROJECTS/"
        "RAG-projects/RAG_finance"
    )).expanduser().resolve()
    out_dir = root / "reports/phase2/quality_improvement/NB22E.lab.v3"
    need(out_dir.is_dir(), "V3_DIRECTORY_NOT_FOUND")

    receipts = {}

    def read_file(relative, expected_sha=None, jsonl=False):
        path = (root / relative).resolve()
        need(path.is_relative_to(root), "PATH_OUTSIDE_PROJECT")
        need(path.is_file(), "FILE_NOT_FOUND: " + relative)
        before = path.stat()
        raw = path.read_bytes()
        after = path.stat()
        need(
            (before.st_size, before.st_mtime_ns)
            == (after.st_size, after.st_mtime_ns),
            "INPUT_CHANGED_DURING_READ: " + relative
        )
        digest = sha(raw)
        if expected_sha is not None:
            need(digest == expected_sha, "HASH_CONFLICT: " + relative)

        text = raw.decode("utf-8")
        value = (
            [parse(line) for line in text.splitlines() if line.strip()]
            if jsonl else parse(text)
        )
        receipts[relative] = {
            "file_bytes_sha256": digest,
            "size_bytes": len(raw),
            "previously_pinned_hash_checked": expected_sha is not None
        }
        return value

    prior_sha = (
        "04fd3d4fc1b034acb9400a6bde709131"
        "219e300cf8c11a9dc1160036d9c4afae"
    )
    prior_path = (
        "reports/phase2/quality_improvement/NB22E.lab.v3/"
        "C018.R08_condition." + prior_sha + ".json"
    )
    prior_condition = read_file(prior_path, prior_sha)

    report_sha = (
        "28178f3b792ae46133d36d3860d1a55f"
        "8e448e4ef69b9200154d5bf0e64ef995"
    )
    results_sha = (
        "aa804f6e34d0243e855d8e92f573bb4b"
        "51d7669720592413c88ef413461d0851"
    )
    report_path = (
        "reports/phase2/quality_improvement/NB22B.v1/"
        "A07.result." + report_sha + ".json"
    )
    results_path = (
        "reports/phase2/quality_improvement/NB22B.v1/"
        "A07.semantic_gap_review_results." + results_sha + ".jsonl"
    )

    report = read_file(report_path, report_sha)
    results = read_file(results_path, results_sha, jsonl=True)

    need(
        report["action"] == 7
        and len(results) == report["response_count"],
        "A07_POPULATION_CONFLICT"
    )
    ids = [row["candidate_id"] for row in results]
    need(
        len(ids) == len(set(ids)) == 1149,
        "A07_IDENTITY_COUNT_CONFLICT"
    )

    def walk(value, pointer=""):
        if isinstance(value, dict):
            yield pointer, value
            for key, child in value.items():
                escaped = str(key).replace("~", "~0").replace("/", "~1")
                yield from walk(child, pointer + "/" + escaped)
        elif isinstance(value, list):
            for index, child in enumerate(value):
                yield from walk(child, pointer + "/" + str(index))

    # کنترل اتصال فایل نتایج به خود گزارش.
    result_references = [
        node for _, node in walk(report)
        if node.get("path") == results_path
    ]
    need(
        any(
            node.get("file_bytes_sha256") == results_sha
            for node in result_references
        ),
        "A07_REPORT_RESULTS_RECEIPT_CONFLICT"
    )

    counts = {}
    pending = []
    prior_adapter_ids = []

    valid_status = "LOCALLY_VALID_MODEL_GAP_REVIEW"
    held_status = "RESPONSE_HELD_NO_RESEND"
    http_status = "HTTP_FAILURE_NO_RESEND"

    for row in results:
        status = row["status"]
        need(isinstance(status, str), "INVALID_A07_STATUS")
        counts[status] = counts.get(status, 0) + 1

        if row.get("offline_format_adapter") is not None:
            prior_adapter_ids.append(row["candidate_id"])

        if status != valid_status:
            pending.append({
                "candidate_id": row["candidate_id"],
                "recorded_status": status,
                "original_record_preserved": row,
                "source_attempt_as_recorded": row.get("source_attempt"),
                "raw_attempt_binding_established_by_this_cell": False,
                "missing_reference_proves_missing_raw_response": False,
                "new_api_request_created": False
            })

    need(
        counts.get(valid_status, 0)
        == report["valid_model_gap_reviews_total"],
        "A07_VALID_COUNT_CONFLICT"
    )

    # حفظ شمارهای گزارش؛ اگر دو بخش جمعیت متفاوت دارند، یکی جای دیگری نمی‌نشیند.
    report_status_counts = report.get("semantic_review_result_status_counts")
    if report_status_counts is not None:
        need(
            report_status_counts == counts,
            "A07_RESULT_STATUS_COUNTS_CONFLICT"
        )

    notebook_path = (
        "notebooks/phase2/22b_quality_improvement_and_acceptance.ipynb"
    )
    notebook = read_file(notebook_path)
    need(
        notebook.get("nbformat") == 4
        and isinstance(notebook.get("cells"), list),
        "NB22B_NOTEBOOK_FORMAT_CONFLICT"
    )

    # فقط متن کد؛ خروجی‌های حجیم و هیچ کدی اجرا نمی‌شوند.
    producer_cells = []
    for index, cell in enumerate(notebook["cells"]):
        if cell.get("cell_type") != "code":
            continue
        source = cell.get("source", [])
        text = source if isinstance(source, str) else "".join(source)
        producer_cells.append({
            "cell_index": index,
            "cell_id": cell.get("id"),
            "source_utf8_sha256": sha(text.encode("utf-8")),
            "source": text
        })

    package = {
        "schema_version": "NB22E_V3_C019_R08_EXISTING_INPUTS_V1",
        "status": "R08_EXISTING_A07_AND_PRODUCER_BOUND_FOR_RECOVERY_REVIEW",
        "roadmap_action": "R08",
        "input_receipts": receipts,
        "scope_authorization": {
            "basis": "CURRENT_USER_INSTRUCTION_TO_PROCEED_WITH_R08",
            "prior_C018_scope": prior_condition["condition_review"],
            "prior_C018_file_modified": False,
            "R08_independent_followup_now_active": True
        },
        "A07_report": report,
        "A07_results_preserved": results,
        "recorded_status_counts": counts,
        "pending_records": pending,
        "pending_record_count": len(pending),
        "received_held_count_as_recorded": counts.get(held_status, 0),
        "http_failure_count_as_recorded": counts.get(http_status, 0),
        "http_failure_outcome_inferred": False,
        "prior_adapter_candidate_ids": prior_adapter_ids,
        "prior_adapter_presence_proves_semantic_correctness": False,
        "producer_notebook_path": notebook_path,
        "producer_code_cells": producer_cells,
        "producer_notebook_executed": False,
        "subtasks": {
            "R08.1": {
                "status": "PARTIAL_RECORDED_STATES_COUNTED_RAW_OUTCOME_REVIEW_OPEN",
                "basis_fa": (
                    "وضعیت‌های ثبت‌شده شمارش شدند؛ "
                    "علت دقیق Held و نتیجهٔ Attemptهای HTTP هنوز بررسی محتوایی می‌خواهند."
                )
            },
            "R08.2": {
                "status": "OPEN_ATTEMPT_REFERENCES_PRESERVED",
                "basis_fa": (
                    "ارجاع‌های موجود حفظ شدند؛ "
                    "اتصال پاسخ خام و ورودی دقیق هنوز احراز نشده است."
                )
            },
            "R08.3": {"status": "OPEN_ACTUAL_VALIDATOR_REVIEW_REQUIRED"},
            "R08.4": {
                "status": "OPEN_REUSE_PRIOR_ADAPTER_BEFORE_ANY_NEW_ADAPTER"
            },
            "R08.5": {
                "status": "OPEN_FINAL_RECOVERY_ACCOUNTING_NOT_YET_ESTABLISHED"
            }
        },
        "raw_responses_modified": False,
        "historical_results_modified": False,
        "successful_reviews_repeated": False,
        "new_api_calls": 0,
        "resends": 0,
        "new_semantic_judgments": 0,
        "newly_recovered_records": 0,
        "metrics_calculated": False,
        "all_R08_items_complete": False,
        "advance_to_next_roadmap_action": False,
        "next_step": "READ_BOUND_PRODUCER_AND_PENDING_RECORDS_FOR_EXACT_RAW_RECOVERY",
        "user_action_fa": (
            "فقط فایل JSON معرفی‌شده در output را در همین چت پیوست کنید؛ "
            "این بسته شامل نتیجهٔ موجود و کد Producer است."
        )
    }

    data = (
        json.dumps(
            package, ensure_ascii=False, sort_keys=True,
            indent=2, allow_nan=False
        ) + "\n"
    ).encode("utf-8")
    digest = sha(data)
    output_path = out_dir / (
        "C019.R08_existing_inputs." + digest + ".json"
    )

    if output_path.exists():
        need(output_path.read_bytes() == data, "OUTPUT_CONTENT_CONFLICT")
    else:
        with output_path.open("xb") as handle:
            handle.write(data)

    saved = output_path.read_bytes()
    need(sha(saved) == digest, "OUTPUT_HASH_CONFLICT")
    need(parse(saved.decode("utf-8")) == package, "OUTPUT_RELOAD_CONFLICT")

    print(json.dumps({
        "status": package["status"],
        "roadmap_action": "R08",
        "response_count": len(results),
        "recorded_status_counts": counts,
        "pending_record_count": len(pending),
        "prior_adapter_record_count": len(prior_adapter_ids),
        "producer_code_cell_count": len(producer_cells),
        "new_api_calls": 0,
        "resends": 0,
        "newly_recovered_records": 0,
        "all_R08_items_complete": False,
        "advance_to_next_roadmap_action": False,
        "write_status": "WRITTEN",
        "output": {
            "path": str(output_path.relative_to(root)),
            "file_bytes_sha256": digest,
            "size_bytes": len(data),
            "reload_validation": "PASS"
        },
        "user_action_fa": package["user_action_fa"]
    }, ensure_ascii=False, indent=2))

    return package


V3_C019 = run_v3_c019()
````

````python
# C020 — R08: اتصال Attemptهای واقعی و بازآزمایی Validator قبلی
# بدون import، API، resend یا تغییر فایل‌های تاریخی

def run_v3_c020():
    missing = [
        name for name in ("Path", "os", "json", "hashlib")
        if name not in globals()
    ]
    if missing:
        raise RuntimeError("ابتدا سل imports را اجرا کنید: " + ", ".join(missing))

    class RecoveryCheckError(Exception):
        pass

    def need(condition, message):
        if not condition:
            raise RecoveryCheckError(message)

    def sha(raw):
        return hashlib.sha256(raw).hexdigest()

    def canonical(value):
        return json.dumps(
            value, ensure_ascii=False, sort_keys=True,
            separators=(",", ":"), allow_nan=False
        ).encode("utf-8")

    def unique_object(items):
        result = {}
        for key, value in items:
            need(key not in result, "DUPLICATE_JSON_KEY: " + key)
            result[key] = value
        return result

    def reject_constant(value):
        raise ValueError("INVALID_JSON_CONSTANT: " + value)

    def parse(raw):
        return json.loads(
            raw.decode("utf-8"),
            object_pairs_hook=unique_object,
            parse_constant=reject_constant
        )

    root = Path(os.environ.get(
        "RAG_FINANCE_ROOT",
        "/Users/macbookpro/AMIR_DATA/00_WORKS/01_PROJECTS/"
        "RAG-projects/RAG_finance"
    )).expanduser().resolve()
    out_dir = root / "reports/phase2/quality_improvement/NB22E.lab.v3"

    parent_sha = (
        "a4b58b0ab44a206c959fcfe0c6f06e640"
        "ac92af207509be7a95cfbc292029ced"
    )
    parent_path = out_dir / (
        "C019.R08_existing_inputs." + parent_sha + ".json"
    )
    parent_raw = parent_path.read_bytes()
    need(sha(parent_raw) == parent_sha, "C019_HASH_CONFLICT")
    parent = parse(parent_raw)

    pending = {
        row["candidate_id"]: row
        for row in parent["pending_records"]
    }
    need(len(pending) == parent["pending_record_count"] == 262,
         "PENDING_POPULATION_CONFLICT")

    # فقط تابع Validator بررسی‌شده تعریف می‌شود؛
    # هیچ سل یا تابع اجراکنندهٔ نوت‌بوک قبلی اجرا نمی‌شود.
    source_hash = (
        "a8d53287e8db8336d1b0d6c0dad6ec5bb"
        "37819112cacf8e9dd1215d16bfe6279"
    )
    matches = [
        cell for cell in parent["producer_code_cells"]
        if cell["source_utf8_sha256"] == source_hash
    ]
    need(len(matches) == 1, "VALIDATOR_SOURCE_NOT_UNIQUE")
    source = matches[0]["source"]
    need(sha(source.encode("utf-8")) == source_hash,
         "VALIDATOR_SOURCE_HASH_CONFLICT")
    start = source.index("def validate_gap_review(data, result):")
    end = source.index("\ndef gap_request(data):", start)
    validator_source = source[start:end]

    schema_keys = (
        "candidate_id", "response_document_sha256",
        "requirement_coverage", "citation_links",
        "additional_factual_units", "attribution_assertions",
        "factual_extraction_review",
        "reference_requirement_qualification", "reason_fa"
    )
    namespace = {
        "need": need,
        "EvidenceGap": RecoveryCheckError,
        "GAP_REVIEW_SCHEMA": dict.fromkeys(schema_keys)
    }
    exec(
        compile(validator_source, "<pinned-gap-validator>", "exec"),
        namespace
    )
    validate = namespace["validate_gap_review"]

    jobs_root = root / (
        "artifacts/phase2/quality_improvement/"
        "NB22B.semantic_gap_review_v1/jobs"
    )
    need(jobs_root.is_dir(), "EXACT_JOBS_DIRECTORY_NOT_FOUND")

    def read_stable(path):
        resolved = path.resolve()
        need(resolved.is_relative_to(root), "PATH_OUTSIDE_PROJECT")
        before = resolved.stat()
        raw = resolved.read_bytes()
        after = resolved.stat()
        need(
            (before.st_size, before.st_mtime_ns)
            == (after.st_size, after.st_mtime_ns),
            "FILE_CHANGED_DURING_READ"
        )
        return raw

    grouped = {candidate_id: [] for candidate_id in pending}
    discovery_issues = []

    # فقط input.json در همان مسیر تولیدکننده؛ بدون جست‌وجوی کل پروژه.
    for directory in sorted(jobs_root.iterdir()):
        if not directory.is_dir():
            continue
        input_path = directory / "input.json"
        if not input_path.is_file():
            continue
        try:
            raw = read_stable(input_path)
            data = parse(raw)
            candidate_id = data.get("candidate_id")
            if candidate_id in pending:
                grouped[candidate_id].append(directory)
        except (OSError, ValueError, TypeError, AttributeError,
                RecoveryCheckError) as error:
            discovery_issues.append({
                "path": str(input_path.relative_to(root)),
                "error": str(error),
                "candidate_identity_not_established": True
            })

    attempt_reviews = []

    def inspect_attempt(candidate_id, directory):
        record = {
            "candidate_id": candidate_id,
            "directory": str(directory.relative_to(root)),
            "file_receipts": {},
            "raw_response_modified": False,
            "semantic_content_changed": False
        }

        def load(name, decode=True):
            path = directory / name
            raw = read_stable(path)
            record["file_receipts"][name] = sha(raw)
            return parse(raw) if decode else raw

        try:
            data = load("input.json")
            binding = load("binding.json")
            request_raw = load("request.json", decode=False)
            request = parse(request_raw)
            key = directory.name

            need(data["candidate_id"] == candidate_id, "INPUT_ID_CONFLICT")
            need(sha(canonical(binding)) == key, "JOB_KEY_CONFLICT")
            need(
                binding["input_sha256"] == sha(canonical(data)),
                "INPUT_BINDING_CONFLICT"
            )
            need(
                binding["request_sha256"] == sha(request_raw)
                == sha(canonical(request)),
                "REQUEST_BINDING_CONFLICT"
            )
            need(
                binding["task_version"] == data["task_version"]
                == "NB22B_HISTORICAL_SEMANTIC_GAPS_V1",
                "TASK_VERSION_CONFLICT"
            )

            messages = request["messages"]
            users = [
                message["content"] for message in messages
                if message.get("role") == "user"
            ]
            need(len(users) == 1, "REQUEST_USER_MESSAGE_CONFLICT")
            payload_text, separator, _ = users[0].partition(
                "\nFinal instruction:"
            )
            need(bool(separator), "REQUEST_ENVELOPE_CONFLICT")
            payload = parse(payload_text.encode("utf-8"))
            need(payload["input"] == data, "REQUEST_INPUT_CONTENT_CONFLICT")

            record["input"] = data
            record["binding"] = binding
            record["request_input_binding_verified"] = True

            if not (directory / "attempt.json").is_file():
                record["status"] = "NO_ATTEMPT_RECORD"
                return record

            attempt = load("attempt.json")
            need(
                attempt["job_key"] == key
                and attempt["request_sha256"] == binding["request_sha256"],
                "ATTEMPT_BINDING_CONFLICT"
            )

            if not (
                (directory / "receipt.json").is_file()
                and (directory / "response.body").is_file()
            ):
                record["status"] = "OUTCOME_UNKNOWN_MISSING_BODY_OR_RECEIPT"
                return record

            receipt = load("receipt.json")
            body_raw = load("response.body", decode=False)
            need(
                receipt["request_sha256"] == binding["request_sha256"]
                and receipt["body_sha256"] == sha(body_raw),
                "RAW_RESPONSE_RECEIPT_CONFLICT"
            )
            record["receipt"] = receipt
            record["raw_response_receipt_verified"] = True

            if receipt["http_status"] != 200:
                record["status"] = "HTTP_FAILURE_RECEIVED"
                record["http_status"] = receipt["http_status"]
                record["error_body_text"] = body_raw.decode(
                    "utf-8", errors="replace"
                )
                return record

            body = parse(body_raw)
            model = body.get("model")
            need(
                isinstance(model, str)
                and model.rsplit("/", 1)[-1].lower() in {
                    "deepseek-v4-flash", "deepseek-v4-flash-0731"
                },
                "RESPONSE_MODEL_MISMATCH"
            )
            choices = body.get("choices")
            need(
                isinstance(choices, list) and len(choices) == 1,
                "CHAT_CHOICES_SCHEMA"
            )
            choice = choices[0]
            record["response_model"] = model
            record["finish_reason"] = choice.get("finish_reason")
            record["usage"] = body.get("usage")
            content = choice["message"].get("content")
            record["raw_assistant_content"] = content

            if choice.get("finish_reason") != "stop":
                record["status"] = "STILL_HELD_INCOMPLETE_OUTPUT"
                return record

            need(isinstance(content, str), "MODEL_CONTENT_TYPE")
            text = content.strip()
            fence_adapter = None
            if text.startswith("```json\n") and text.endswith("\n```"):
                text = text[8:-4].strip()
                fence_adapter = "EXISTING_SINGLE_OUTER_JSON_FENCE"
            elif text.startswith("```\n") and text.endswith("\n```"):
                text = text[4:-4].strip()
                fence_adapter = "EXISTING_SINGLE_OUTER_FENCE"

            decoded = parse(text.encode("utf-8"))
            record["decoded_model_output"] = decoded
            checked = validate(data, decoded)
            record["status"] = "RECOVERED_BY_EXISTING_VALIDATOR"
            record["validated_review"] = checked
            record["fence_adapter"] = fence_adapter
            record["semantic_correctness_independently_verified"] = False
            return record

        except (OSError, ValueError, KeyError, TypeError,
                AttributeError, IndexError, RecoveryCheckError) as error:
            record["status"] = (
                "STILL_HELD_VALIDATOR_REJECTION"
                if record.get("raw_response_receipt_verified")
                else "BINDING_OR_ACCESS_UNRESOLVED"
            )
            record["reason"] = (
                str(error) if isinstance(error, RecoveryCheckError)
                else type(error).__name__ + ": " + str(error)
            )
            return record

    dispositions = []
    for candidate_id in sorted(pending):
        attempts = [
            inspect_attempt(candidate_id, directory)
            for directory in grouped[candidate_id]
        ]
        attempt_reviews.extend(attempts)
        valid = [
            record for record in attempts
            if record["status"] == "RECOVERED_BY_EXISTING_VALIDATOR"
        ]
        distinct_valid = {
            sha(canonical(record["validated_review"]["result"]))
            for record in valid
        }

        if len(distinct_valid) == 1:
            status = "RECOVERED_LOCALLY_NOT_SEMANTICALLY_APPROVED"
        elif len(distinct_valid) > 1:
            status = "CONFLICTING_VALID_ATTEMPTS_PRESERVED"
        elif not attempts:
            status = "NO_BOUND_JOB_FOUND_IN_DECLARED_DIRECTORY"
        else:
            states = {record["status"] for record in attempts}
            if states == {"HTTP_FAILURE_RECEIVED"}:
                status = "HTTP_FAILURE_RECEIVED_NO_RESEND"
            elif states.issubset({
                "STILL_HELD_INCOMPLETE_OUTPUT",
                "STILL_HELD_VALIDATOR_REJECTION"
            }):
                status = "STILL_HELD"
            else:
                status = "ATTEMPT_BINDING_OR_OUTCOME_REQUIRES_REVIEW"

        dispositions.append({
            "candidate_id": candidate_id,
            "original_status": pending[candidate_id]["recorded_status"],
            "status": status,
            "attempt_directories": [
                record["directory"] for record in attempts
            ],
            "valid_attempt_count": len(valid),
            "distinct_valid_result_count": len(distinct_valid),
            "latest_attempt_selected": False
        })

    counts = {}
    for row in dispositions:
        counts[row["status"]] = counts.get(row["status"], 0) + 1
    need(sum(counts.values()) == len(pending), "FINAL_ACCOUNTING_CONFLICT")

    result = {
        "schema_version": "NB22E_V3_C020_R08_V1",
        "status": "R08_RAW_ATTEMPTS_BOUND_EXISTING_VALIDATOR_REPLAYED",
        "roadmap_action": "R08",
        "parent_file_bytes_sha256": parent_sha,
        "validator_source_sha256": source_hash,
        "validator_function": "validate_gap_review",
        "new_adapter_rules_introduced": False,
        "previous_valid_reviews_preserved": 887,
        "previous_valid_reviews_revalidated": False,
        "pending_population": len(pending),
        "disposition_counts": counts,
        "dispositions": dispositions,
        "attempt_reviews": attempt_reviews,
        "job_input_read_issues": discovery_issues,
        "newly_recovered_records": counts.get(
            "RECOVERED_LOCALLY_NOT_SEMANTICALLY_APPROVED", 0
        ),
        "original_files_modified": False,
        "partial_output_completed_or_repaired": False,
        "quotes_or_verdicts_changed": False,
        "human_gold_claimed": False,
        "new_api_calls": 0,
        "resends": 0,
        "new_semantic_judgments": 0,
        "metrics_calculated": False,
        "all_R08_items_complete": False,
        "advance_to_next_roadmap_action": False,
        "next_step": "REVIEW_BOUND_RAW_FAILURES_FOR_ONLY_JUSTIFIED_MECHANICAL_RECOVERY",
        "user_action_fa": (
            "فایل JSON خروجی را پیوست کنید تا ساختار واقعی "
            "خطاهای باقی‌مانده بررسی شود؛ پاسخ خام دوباره استخراج نمی‌شود."
        )
    }

    raw = (
        json.dumps(
            result, ensure_ascii=False, sort_keys=True,
            indent=2, allow_nan=False
        ) + "\n"
    ).encode("utf-8")
    digest = sha(raw)
    target = out_dir / ("C020.R08_raw_recovery." + digest + ".json")
    if target.exists():
        need(target.read_bytes() == raw, "OUTPUT_CONTENT_CONFLICT")
    else:
        with target.open("xb") as handle:
            handle.write(raw)

    saved = target.read_bytes()
    need(sha(saved) == digest, "OUTPUT_HASH_CONFLICT")
    need(parse(saved) == result, "OUTPUT_RELOAD_CONFLICT")

    print(json.dumps({
        "status": result["status"],
        "roadmap_action": "R08",
        "previous_valid_reviews_preserved": 887,
        "pending_population": len(pending),
        "attempts_inspected": len(attempt_reviews),
        "disposition_counts": counts,
        "newly_recovered_records": result["newly_recovered_records"],
        "job_input_read_issue_count": len(discovery_issues),
        "new_api_calls": 0,
        "resends": 0,
        "all_R08_items_complete": False,
        "write_status": "WRITTEN",
        "output": {
            "path": str(target.relative_to(root)),
            "file_bytes_sha256": digest,
            "size_bytes": len(raw),
            "reload_validation": "PASS"
        },
        "user_action_fa": result["user_action_fa"]
    }, ensure_ascii=False, indent=2))
    return result


V3_C020 = run_v3_c020()
````

  و بازگرداندن نقل‌قول دقیق از متن اصلی؛

````python
# C021 — R08: Adapter محدود، Validator تاریخی و حسابداری نهایی
# بدون import، API، resend یا تغییر فایل‌های تاریخی

def run_v3_c021():
    missing = [
        name for name in ("Path", "os", "json", "hashlib")
        if name not in globals()
    ]
    if missing:
        raise RuntimeError("ابتدا سل imports را اجرا کنید: " + ", ".join(missing))

    class CheckError(Exception):
        pass

    def need(ok, message):
        if not ok:
            raise CheckError(message)

    def sha(raw):
        return hashlib.sha256(raw).hexdigest()

    def canonical(value):
        return json.dumps(
            value, ensure_ascii=False, sort_keys=True,
            separators=(",", ":"), allow_nan=False
        ).encode("utf-8")

    def unique_object(items):
        result = {}
        for key, value in items:
            need(key not in result, "DUPLICATE_JSON_KEY: " + key)
            result[key] = value
        return result

    def reject_constant(value):
        raise ValueError("INVALID_JSON_CONSTANT: " + value)

    def parse(raw):
        return json.loads(
            raw.decode("utf-8"),
            object_pairs_hook=unique_object,
            parse_constant=reject_constant
        )

    root = Path(os.environ.get(
        "RAG_FINANCE_ROOT",
        "/Users/macbookpro/AMIR_DATA/00_WORKS/01_PROJECTS/"
        "RAG-projects/RAG_finance"
    )).expanduser().resolve()
    out_dir = root / "reports/phase2/quality_improvement/NB22E.lab.v3"
    receipts = {}

    def read_pinned(prefix, digest):
        path = out_dir / (prefix + digest + ".json")
        raw = path.read_bytes()
        need(sha(raw) == digest, "HASH_CONFLICT: " + str(path))
        receipts[prefix] = {
            "path": str(path.relative_to(root)),
            "file_bytes_sha256": digest
        }
        return parse(raw)

    source = read_pinned(
        "C020.R08_raw_recovery.",
        "3906f065e476222f531aba865159213898c9f5523177ce1e71a94709e188cc1b"
    )
    original_package = read_pinned(
        "C019.R08_existing_inputs.",
        "a4b58b0ab44a206c959fcfe0c6f06e640ac92af207509be7a95cfbc292029ced"
    )

    need(
        source["pending_population"] == 262
        and source["previous_valid_reviews_preserved"] == 887
        and not source["job_input_read_issues"],
        "SOURCE_ACCOUNTING_CONFLICT"
    )
    attempts = source["attempt_reviews"]
    ids = [item["candidate_id"] for item in attempts]
    need(len(ids) == len(set(ids)) == 262, "ATTEMPT_IDENTITY_CONFLICT")

    # اثبات اینکه فایل‌های خام از زمان اتصال C020 تغییر نکرده‌اند.
    for attempt in attempts:
        directory = (root / attempt["directory"]).resolve()
        need(directory.is_relative_to(root), "PATH_OUTSIDE_PROJECT")
        need(
            attempt.get("request_input_binding_verified") is True
            and attempt.get("raw_response_receipt_verified") is True,
            attempt["candidate_id"] + ": INCOMPLETE_RAW_BINDING"
        )
        for name, expected in attempt["file_receipts"].items():
            path = (directory / name).resolve()
            need(path.parent == directory, "UNEXPECTED_ATTEMPT_FILENAME")
            need(
                path.is_file() and sha(path.read_bytes()) == expected,
                attempt["candidate_id"] + ": RAW_FILE_CHANGED:" + name
            )

    validator_sha = (
        "a8d53287e8db8336d1b0d6c0dad6ec5bb"
        "37819112cacf8e9dd1215d16bfe6279"
    )
    matches = [
        cell for cell in original_package["producer_code_cells"]
        if cell["source_utf8_sha256"] == validator_sha
    ]
    need(len(matches) == 1, "VALIDATOR_SOURCE_NOT_UNIQUE")
    code = matches[0]["source"]
    need(sha(code.encode("utf-8")) == validator_sha, "VALIDATOR_HASH_CONFLICT")

    start = code.index("def validate_gap_review(data, result):")
    end = code.index("\ndef gap_request(data):", start)
    keys = (
        "candidate_id", "response_document_sha256",
        "requirement_coverage", "citation_links",
        "additional_factual_units", "attribution_assertions",
        "factual_extraction_review",
        "reference_requirement_qualification", "reason_fa"
    )
    namespace = {
        "need": need,
        "GAP_REVIEW_SCHEMA": dict.fromkeys(keys)
    }
    # فقط تعریف تابع Validator خوانده‌شده؛ هیچ Producer یا dispatch اجرا نمی‌شود.
    exec(
        compile(code[start:end], "<pinned-gap-validator>", "exec"),
        namespace
    )
    validate = namespace["validate_gap_review"]

    adapter_version = "R08_LIMITED_FORMAT_AND_UNIQUE_QUOTE_ADAPTER_V1"
    administrative_note = (
        "یادداشت Adapter: صرفاً بازیابی مکانیکی پاسخ ذخیره‌شده؛ "
        "صحت معنایی، استقلال و صلاحیت مرجع احراز نشده است."
    )

    def normalize_with_positions(text):
        output = []
        positions = []
        substitutions = {"ي": "ی", "ك": "ک", "\u200c": " "}
        for index, character in enumerate(text):
            character = substitutions.get(character, character)
            if character.isspace():
                if not output or output[-1] == " ":
                    continue
                character = " "
            output.append(character)
            positions.append(index)
        if output and output[-1] == " ":
            output.pop()
            positions.pop()
        return "".join(output), positions

    def exact_unique_quote(quote, text):
        if not isinstance(quote, str) or not quote or quote in text:
            return quote
        normalized_quote, _ = normalize_with_positions(quote)
        normalized_text, positions = normalize_with_positions(text)
        if not normalized_quote:
            return quote

        starts = []
        cursor = 0
        while True:
            found = normalized_text.find(normalized_quote, cursor)
            if found < 0:
                break
            starts.append(found)
            cursor = found + 1
        if len(starts) != 1:
            return quote

        found = starts[0]
        replacement = text[
            positions[found]:
            positions[found + len(normalized_quote) - 1] + 1
        ]
        return (
            replacement
            if normalize_with_positions(replacement)[0] == normalized_quote
            else quote
        )

    def adapt(attempt):
        data = attempt["input"]
        result = parse(canonical(attempt["decoded_model_output"]))
        changes = []

        def replace(container, key, value, pointer, rule):
            old = container[key]
            container[key] = value
            changes.append({
                "pointer": pointer,
                "rule": rule,
                "old_value": old,
                "new_value": value
            })

        if "reason_fa" not in result:
            result["reason_fa"] = administrative_note
            changes.append({
                "pointer": "/reason_fa",
                "rule": "GENERATED_ADMINISTRATIVE_LIMITATION_NOT_MODEL_REASON",
                "field_previously_present": False,
                "new_value": administrative_note
            })

        removable_null_fields = {
            "requirement_coverage": {
                "reason_fa_note", "requirement_index_note"
            },
            "citation_links": {"reason_fa_note"},
            "additional_factual_units": {
                "reason_fa_note", "claim_index_note", "role"
            }
        }
        for collection, allowed in removable_null_fields.items():
            for index, item in enumerate(result.get(collection, [])):
                for key in sorted(allowed):
                    if key in item and item[key] is None:
                        del item[key]
                        changes.append({
                            "pointer": "/" + collection + "/" + str(index) + "/" + key,
                            "rule": "REMOVE_EXPLICIT_EXTRA_NULL_METADATA",
                            "old_value": None,
                            "field_removed": True
                        })

        evidence = {
            item["evidence_id"]: item["text"]
            for item in data["evidence"]
        }

        for collection, quote_key in (
            ("requirement_coverage", "answer_quotes"),
            ("citation_links", "evidence_quotes")
        ):
            for index, item in enumerate(result.get(collection, [])):
                text = (
                    data["answer_text"]
                    if collection == "requirement_coverage"
                    else evidence.get(item.get("evidence_id"))
                )
                if text is None:
                    continue
                for quote_index, quote in enumerate(item.get(quote_key, [])):
                    replacement = exact_unique_quote(quote, text)
                    if replacement != quote:
                        replace(
                            item[quote_key], quote_index, replacement,
                            "/" + collection + "/" + str(index)
                            + "/" + quote_key + "/" + str(quote_index),
                            "UNIQUE_LIMITED_NORMALIZATION_TO_EXACT_SOURCE_SPAN"
                        )

        for index, item in enumerate(result.get("additional_factual_units", [])):
            quote = item.get("answer_quote")
            replacement = exact_unique_quote(quote, data["answer_text"])
            if replacement != quote:
                replace(
                    item, "answer_quote", replacement,
                    "/additional_factual_units/" + str(index) + "/answer_quote",
                    "UNIQUE_LIMITED_NORMALIZATION_TO_EXACT_ANSWER_SPAN"
                )

            for quote_index, witness in enumerate(item.get("evidence_quotes", [])):
                pointer = (
                    "/additional_factual_units/" + str(index)
                    + "/evidence_quotes/" + str(quote_index)
                )
                if isinstance(witness, str):
                    matches = []
                    for evidence_id in item.get("evidence_ids", []):
                        if evidence_id not in evidence:
                            continue
                        replacement = exact_unique_quote(
                            witness, evidence[evidence_id]
                        )
                        if (
                            witness in evidence[evidence_id]
                            or replacement != witness
                        ):
                            matches.append((evidence_id, replacement))
                    if len(matches) == 1:
                        evidence_id, replacement = matches[0]
                        replace(
                            item["evidence_quotes"], quote_index,
                            {"evidence_id": evidence_id, "quote": replacement},
                            pointer,
                            "STRING_WITNESS_TO_UNIQUE_ALLOWED_SOURCE"
                        )
                elif (
                    isinstance(witness, dict)
                    and witness.get("evidence_id") in evidence
                ):
                    quote = witness.get("quote")
                    replacement = exact_unique_quote(
                        quote, evidence[witness["evidence_id"]]
                    )
                    if replacement != quote:
                        replace(
                            witness, "quote", replacement, pointer + "/quote",
                            "UNIQUE_LIMITED_NORMALIZATION_TO_EXACT_SOURCE_SPAN"
                        )

        for index, item in enumerate(result.get("attribution_assertions", [])):
            quote = item.get("answer_quote")
            replacement = exact_unique_quote(quote, data["answer_text"])
            if replacement != quote:
                replace(
                    item, "answer_quote", replacement,
                    "/attribution_assertions/" + str(index) + "/answer_quote",
                    "UNIQUE_LIMITED_NORMALIZATION_TO_EXACT_ANSWER_SPAN"
                )
            evidence_id = item.get("reference_evidence_id")
            if evidence_id in evidence:
                quote = item.get("reference_quote")
                replacement = exact_unique_quote(quote, evidence[evidence_id])
                if replacement != quote:
                    replace(
                        item, "reference_quote", replacement,
                        "/attribution_assertions/" + str(index) + "/reference_quote",
                        "UNIQUE_LIMITED_NORMALIZATION_TO_EXACT_SOURCE_SPAN"
                    )

        return result, changes

    residual_categories = {
        "GAP_REVIEW_RESPONSE_IDENTITY": "IDENTITY_CONFLICT_NO_ID_REPAIR",
        "REQUIREMENT_EXACT_COVERAGE": "MISSING_OR_CONFLICTING_UNIT_IDENTITIES",
        "LINK_SUPPORT_ENUM": "INVALID_ENUM_NO_VERDICT_CONVERSION",
        "LINK_QUOTE_WRONG_EVIDENCE": "NO_APPROVED_UNIQUE_QUOTE_MATCH",
        "REQUIREMENT_ANSWER_QUOTE": "NO_APPROVED_UNIQUE_QUOTE_MATCH",
        "ATTRIBUTION_SOURCE_WITNESS": "NO_APPROVED_UNIQUE_QUOTE_MATCH",
        "ATTRIBUTION_ANSWER_WITNESS": "NO_APPROVED_UNIQUE_QUOTE_MATCH",
        "FACTUAL_WITNESS_NOT_FOUND": "NO_APPROVED_UNIQUE_QUOTE_MATCH"
    }

    outcomes = []
    for attempt in attempts:
        outcome = {
            "candidate_id": attempt["candidate_id"],
            "source_directory": attempt["directory"],
            "source_file_receipts": attempt["file_receipts"],
            "original_status": attempt["status"],
            "original_reason": attempt.get("reason"),
            "raw_response_modified": False,
            "semantic_verdicts_changed": False,
            "formal_approval_granted": False
        }

        if attempt["status"] == "HTTP_FAILURE_RECEIVED":
            outcome.update({
                "status": "HTTP_FAILURE_RECEIVED_NO_RESEND",
                "category": "RECEIVED_NON200_HTTP_RESPONSE",
                "http_status": attempt["http_status"]
            })
        elif attempt["status"] == "STILL_HELD_INCOMPLETE_OUTPUT":
            outcome.update({
                "status": "STILL_HELD",
                "category": "INCOMPLETE_OUTPUT_NOT_REPAIRED",
                "finish_reason": attempt["finish_reason"]
            })
        else:
            need(
                attempt["status"] == "STILL_HELD_VALIDATOR_REJECTION",
                "UNEXPECTED_ATTEMPT_STATE"
            )
            adapted, changes = adapt(attempt)
            outcome["original_model_output_sha256"] = sha(
                canonical(attempt["decoded_model_output"])
            )
            outcome["adapted_output_sha256"] = sha(canonical(adapted))
            outcome["changes"] = changes
            outcome["adapted_output"] = adapted
            outcome["administrative_reason_generated"] = (
                "reason_fa" not in attempt["decoded_model_output"]
            )
            try:
                checked = validate(attempt["input"], adapted)
                outcome.update({
                    "status": "RECOVERED_LOCALLY",
                    "category": "ADAPTED_OUTPUT_PASSES_UNCHANGED_VALIDATOR",
                    "validated_review": checked,
                    "semantic_correctness_independently_verified": False
                })
            except (CheckError, ValueError, KeyError, TypeError,
                    AttributeError, IndexError) as error:
                reason = (
                    str(error) if isinstance(error, CheckError)
                    else type(error).__name__ + ": " + str(error)
                )
                outcome.update({
                    "status": "STILL_HELD",
                    "residual_validator_reason": reason,
                    "category": residual_categories.get(
                        reason, "NO_APPROVED_SCHEMA_ADAPTER_RULE"
                    )
                })
        outcomes.append(outcome)

    counts = {}
    categories = {}
    for outcome in outcomes:
        status = outcome["status"]
        category = outcome["category"]
        counts[status] = counts.get(status, 0) + 1
        categories[category] = categories.get(category, 0) + 1

    need(
        counts == {
            "RECOVERED_LOCALLY": 160,
            "STILL_HELD": 101,
            "HTTP_FAILURE_RECEIVED_NO_RESEND": 1
        },
        "TESTED_ADAPTER_OUTCOME_CONFLICT: " + str(counts)
    )
    need(sum(counts.values()) == 262, "PENDING_ACCOUNTING_CONFLICT")
    total_valid = 887 + counts["RECOVERED_LOCALLY"]
    need(
        total_valid + counts["STILL_HELD"]
        + counts["HTTP_FAILURE_RECEIVED_NO_RESEND"] == 1149,
        "TOTAL_ACCOUNTING_CONFLICT"
    )

    result = {
        "schema_version": "NB22E_V3_C021_R08_V1",
        "status": "R08_RECOVERY_ASSESSMENT_COMPLETE_WITH_RESIDUALS",
        "roadmap_action": "R08",
        "input_receipts": receipts,
        "adapter_version": adapter_version,
        "validator_source_sha256": validator_sha,
        "validator_relaxed": False,
        "outcomes": outcomes,
        "pending_population": 262,
        "disposition_counts": counts,
        "category_counts": categories,
        "previous_valid_reviews_preserved": 887,
        "previous_valid_reviews_repeated": False,
        "newly_recovered_records": counts["RECOVERED_LOCALLY"],
        "locally_valid_reviews_after_recovery": total_valid,
        "still_held_count": counts["STILL_HELD"],
        "received_http_failure_count": counts["HTTP_FAILURE_RECEIVED_NO_RESEND"],
        "unknown_outcome_count_in_bound_pending_population": 0,
        "all_1149_records_accounted": True,
        "subtasks": {
            "R08.1": "DONE_RECORDED_FAILURE_STATES_AND_OUTCOMES_SEPARATED",
            "R08.2": "DONE_EXACT_RAW_REQUEST_INPUT_RECEIPT_BINDINGS_REUSED",
            "R08.3": "DONE_RESIDUAL_VALIDATOR_CATEGORIES_RECORDED",
            "R08.4": "DONE_VERSIONED_LIMITED_ADAPTER_APPLIED_WITH_CHANGE_LOG",
            "R08.5": "DONE_RECOVERED_STILL_HELD_HTTP_AND_UNKNOWN_REPORTED"
        },
        "R08_recovery_assessment_complete_in_declared_scope": True,
        "all_held_recovered": False,
        "residuals_resolved_semantically": False,
        "residuals_impossible_to_recover_by_any_method": False,
        "formal_reference_qualified": False,
        "human_gold_claimed": False,
        "semantic_correctness_independently_verified": False,
        "historical_files_modified": False,
        "historical_validator_modified": False,
        "new_api_calls": 0,
        "resends": 0,
        "new_model_pairs": 0,
        "new_semantic_judgments": 0,
        "metrics_calculated": False,
        "gate_changed": False,
        "advance_to_next_roadmap_action": False,
        "next_step": "REPORT_R08_RECOVERY_AND_PRESERVED_RESIDUALS",
        "limitations": [
            "Generated top-level limitation notes are Adapter metadata, not model reasons.",
            "Quote normalization is restricted and requires a unique source match.",
            "Passing the validator is local structural validity, not semantic approval.",
            "Residual records remain available for separately scoped followup."
        ]
    }

    raw = (
        json.dumps(
            result, ensure_ascii=False, sort_keys=True,
            indent=2, allow_nan=False
        ) + "\n"
    ).encode("utf-8")
    digest = sha(raw)
    target = out_dir / (
        "C021.R08_recovery_disposition." + digest + ".json"
    )
    if target.exists():
        need(target.read_bytes() == raw, "OUTPUT_CONTENT_CONFLICT")
    else:
        with target.open("xb") as handle:
            handle.write(raw)

    saved = target.read_bytes()
    need(sha(saved) == digest, "OUTPUT_HASH_CONFLICT")
    need(parse(saved) == result, "OUTPUT_RELOAD_CONFLICT")

    print(json.dumps({
        "status": result["status"],
        "roadmap_action": "R08",
        "previous_valid_reviews_preserved": 887,
        "newly_recovered_records": counts["RECOVERED_LOCALLY"],
        "locally_valid_reviews_after_recovery": total_valid,
        "still_held_count": counts["STILL_HELD"],
        "received_http_failure_count": counts["HTTP_FAILURE_RECEIVED_NO_RESEND"],
        "unknown_outcome_count_in_bound_pending_population": 0,
        "category_counts": categories,
        "all_1149_records_accounted": True,
        "R08_recovery_assessment_complete_in_declared_scope": True,
        "all_held_recovered": False,
        "formal_reference_qualified": False,
        "new_api_calls": 0,
        "resends": 0,
        "advance_to_next_roadmap_action": False,
        "write_status": "WRITTEN",
        "output": {
            "path": str(target.relative_to(root)),
            "file_bytes_sha256": digest,
            "reload_validation": "PASS"
        }
    }, ensure_ascii=False, indent=2))
    return result


V3_C021 = run_v3_c021()
````

````python
# C023 — R09: یک آزمایش اکتشافی با رتبه‌های ذخیره‌شده
# اصلاح اتصال baseline مطابق Producer: candidate_id
# بدون import، API، inference، training یا download

def run_v3_c023():
    required = ("Path", "os", "json", "hashlib", "math", "time")
    missing = [name for name in required if name not in globals()]
    if missing:
        raise RuntimeError(
            "ایمپورت‌های زیر باید در سل اول موجود باشند: "
            + ", ".join(missing)
        )

    started = time.monotonic()

    def need(condition, message):
        if not condition:
            raise ValueError(message)

    def sha(raw):
        return hashlib.sha256(raw).hexdigest()

    def canonical(value):
        return json.dumps(
            value, ensure_ascii=False, sort_keys=True,
            separators=(",", ":"), allow_nan=False
        ).encode("utf-8")

    def unique_object(items):
        result = {}
        for key, value in items:
            need(key not in result, "DUPLICATE_JSON_KEY: " + key)
            result[key] = value
        return result

    def reject_constant(value):
        raise ValueError("INVALID_JSON_CONSTANT: " + value)

    def parse(raw):
        return json.loads(
            raw.decode("utf-8"),
            object_pairs_hook=unique_object,
            parse_constant=reject_constant
        )

    root = Path(os.environ.get(
        "RAG_FINANCE_ROOT",
        "/Users/macbookpro/AMIR_DATA/00_WORKS/01_PROJECTS/"
        "RAG-projects/RAG_finance"
    )).expanduser().resolve()
    v2 = root / "reports/phase2/quality_improvement/NB22E.lab.v2"
    v3 = root / "reports/phase2/quality_improvement/NB22E.lab.v3"
    receipts = {}

    def checked_path(relative):
        path = (root / relative).resolve()
        need(path.is_relative_to(root), "PATH_OUTSIDE_PROJECT")
        need(path.is_file(), "FILE_NOT_FOUND: " + relative)
        return path

    def read_bound(relative, digest, jsonl=False):
        path = checked_path(relative)
        raw = path.read_bytes()
        need(sha(raw) == digest, "HASH_CONFLICT: " + relative)
        receipts[relative] = digest
        return (
            [parse(line) for line in raw.splitlines() if line.strip()]
            if jsonl else parse(raw)
        )

    def read_v3(prefix, digest):
        return read_bound(
            str((v3 / (prefix + digest + ".json")).relative_to(root)),
            digest
        )

    def save(stem, value):
        raw = (
            json.dumps(
                value, ensure_ascii=False, sort_keys=True,
                indent=2, allow_nan=False
            ) + "\n"
        ).encode("utf-8")
        digest = sha(raw)
        path = v3 / (stem + "." + digest + ".json")
        if path.exists():
            need(path.read_bytes() == raw, "OUTPUT_CONTENT_CONFLICT")
        else:
            with path.open("xb") as handle:
                handle.write(raw)
        saved = path.read_bytes()
        need(sha(saved) == digest, "OUTPUT_HASH_CONFLICT")
        need(parse(saved) == value, "OUTPUT_RELOAD_CONFLICT")
        return {
            "path": str(path.relative_to(root)),
            "file_bytes_sha256": digest,
            "reload_validation": "PASS"
        }

    previous = read_v3(
        "C022.R09_condition.",
        "ed4bbdce28eb32dc867f5a87b0593e3feccab377c57d2ce8892a2a6a7bcc3e3f"
    )
    recovery = read_v3(
        "C021.R08_recovery_disposition.",
        "72b075dd2ec734a8182e3d1dbe012f32494f5b78ba994ac543108be704f78cd1"
    )
    package = read_v3(
        "C013.R05_binding_inputs.",
        "4b10aaa2eab0e1110854fe0d2374be401ca6197f1aa7c4c9a902d1ff47ad820c"
    )
    need(previous["roadmap_action"] == "R09", "WRONG_PREVIOUS_ACTION")
    need(
        recovery["R08_recovery_assessment_complete_in_declared_scope"] is True,
        "R08_ASSESSMENT_NOT_COMPLETE"
    )

    def packaged(prefix):
        matches = [
            item for item in package["files"]
            if Path(item["path"]).name.startswith(prefix)
        ]
        need(len(matches) == 1, "ARTIFACT_NOT_UNIQUE: " + prefix)
        item = matches[0]
        content = read_bound(
            item["path"], item["file_bytes_sha256"],
            item["content_type"] == "JSONL_RECORDS"
        )
        need(content == item["content"], "PACKAGE_CONTENT_CONFLICT")
        return content

    source_plan = packaged("C063.full_pool24_plan.")
    stored_queries = packaged("C063.full_pool24_completed_queries.")

    report64_sha = (
        "1296107146ae5b6305fcc97cfda07059"
        "deddbcf44fb91f5159aada87165eff5e"
    )
    report64 = read_bound(
        str((v2 / (
            "C064.full_pool24_rank_diagnosis."
            + report64_sha + ".json"
        )).relative_to(root)),
        report64_sha
    )
    history_receipt = report64["cases_artifact"]
    histories = read_bound(
        history_receipt["path"],
        history_receipt["file_bytes_sha256"],
        jsonl=True
    )

    ids = source_plan["query_ids"]
    need(len(ids) == len(set(ids)) == 24, "COHORT_CONFLICT")
    query_index = {row["query_id"]: row for row in stored_queries}
    history_index = {row["query_id"]: row for row in histories}
    need(
        len(query_index) == len(stored_queries) == 24
        and len(history_index) == len(histories) == 24
        and set(ids) == set(query_index) == set(history_index),
        "QUERY_SET_CONFLICT"
    )

    gold_map = {}
    for query_id in ids:
        row = history_index[query_id]
        chunks = [entry["chunk_id"] for entry in row["gold_rank_details"]]
        need(
            len(chunks) == len(set(chunks)) == row["gold_count"]
            and bool(chunks),
            "GOLD_COUNT_CONFLICT: " + query_id
        )
        gold_map[query_id] = sorted(chunks)
    need(
        sha(canonical(gold_map)) == source_plan["historical_gold_sha256"],
        "HISTORICAL_GOLD_FINGERPRINT_CONFLICT"
    )

    baseline_receipt = source_plan["source_bindings"]["baseline"]
    baseline_path = checked_path(baseline_receipt["path"])
    baseline_index = {}
    fingerprint = hashlib.sha256()
    before = baseline_path.stat()

    # مطابق تابع تاریخی c025_subset:
    # شناسهٔ ردیف baseline در candidate_id ذخیره شده است.
    with baseline_path.open("rb") as handle:
        for line in handle:
            fingerprint.update(line)
            if not line.strip():
                continue
            row = parse(line)
            need(isinstance(row, dict), "BASELINE_ROW_NOT_OBJECT")
            candidate_id = row.get("candidate_id")
            if candidate_id in query_index:
                need(
                    candidate_id not in baseline_index,
                    "DUPLICATE_BASELINE_QUERY: " + candidate_id
                )
                baseline_index[candidate_id] = row

    after = baseline_path.stat()
    need(
        (before.st_size, before.st_mtime_ns, before.st_ino)
        == (after.st_size, after.st_mtime_ns, after.st_ino),
        "BASELINE_CHANGED_DURING_READ"
    )
    need(
        fingerprint.hexdigest() == baseline_receipt["file_bytes_sha256"],
        "BASELINE_HASH_CONFLICT"
    )
    need(
        set(baseline_index) == set(ids),
        "BASELINE_QUERY_SET_CONFLICT: " + json.dumps({
            "expected": len(ids),
            "found": len(baseline_index),
            "missing_candidate_ids": sorted(set(ids) - set(baseline_index))
        }, ensure_ascii=False)
    )
    receipts[baseline_receipt["path"]] = baseline_receipt["file_bytes_sha256"]

    experiment = {
        "version": "BASELINE_TOP10_MEMBERSHIP_BGE_ORDER_V1",
        "roadmap_action": "R09",
        "hypothesis_fa": (
            "حفظ دقیق اعضای Top10 baseline و استفاده از ترتیب BGE "
            "ممکن است nDCG@10 را بدون افت Recall@10 و Hit@10 بهتر کند."
        ),
        "cohort": list(ids),
        "cohort_size": 24,
        "data_scope": "PREVIOUSLY_EXPOSED_DEVELOPMENT24_EXPLORATORY",
        "selection_after_prior_results_seen": True,
        "independent_holdout": False,
        "cutoff": 10,
        "candidate_rule": (
            "Sort baseline top10 by their positions in stored full-pool "
            "BGE order; append unchanged baseline tail."
        ),
        "baseline_record_identity_field": "candidate_id",
        "C063_record_identity_field": "query_id",
        "missing_bge_rank_rule": "BLOCK_QUERY_NO_INFERENCE_NO_IMPUTATION",
        "tie_rule": "BGE positions unique; baseline rank as secondary key",
        "parameter_search": False,
        "candidate_method_count": 1,
        "historical_gold_sha256": source_plan["historical_gold_sha256"],
        "novelty_scope": "REVIEWED_NB22E_V2_EXPERIMENTS_ONLY",
        "different_from_reviewed_prior_methods": [
            "RRF changes eligibility or membership rather than preserving top10.",
            "Top20/Top40 reranking can change which ten items are returned.",
            "Preserving top5 does not preserve the full baseline top10."
        ],
        "success_rule": {
            "all_24_queries_required": True,
            "per_query_recall_and_hit_unchanged": True,
            "macro_ndcg_delta_must_exceed": 1e-12,
            "formal_acceptance_granted_by_this_rule": False
        },
        "stop_rule": "ONE_CANDIDATE_ONLY_NO_AUTOMATIC_EXPANSION",
        "runtime": "LOCAL_CPU_RANK_OPERATIONS_ONLY",
        "paid_cost": 0,
        "new_model_pairs": 0,
        "new_api_calls": 0,
        "onnx_required": False,
        "downloads": False,
        "input_receipts": dict(receipts)
    }

    plan_output = save("C023.R09_experiment_plan", experiment)
    metric_names = ("recall_at_10", "ndcg_at_10", "hit_at_10")

    def measure_top10(order, gold):
        need(bool(gold), "EMPTY_GOLD")
        flags = [int(chunk in gold) for chunk in order[:10]]
        ideal = math.fsum(
            1.0 / math.log2(index + 2)
            for index in range(min(10, len(gold)))
        )
        return {
            "recall_at_10": sum(flags) / len(gold),
            "ndcg_at_10": math.fsum(
                flag / math.log2(index + 2)
                for index, flag in enumerate(flags)
            ) / ideal,
            "hit_at_10": float(any(flags))
        }

    completed = []
    blocked = []

    for query_id in ids:
        baseline_row = baseline_index[query_id]
        need(baseline_row["split"] == "development", "SPLIT_CONFLICT")
        baseline_order = baseline_row["ranked_chunk_ids"]
        bge_order = query_index[query_id]["bge_order"]
        need(
            len(baseline_order) >= 10
            and len(baseline_order) == len(set(baseline_order))
            and len(bge_order) == len(set(bge_order)),
            "RANK_IDENTITY_CONFLICT: " + query_id
        )

        bge_rank = {
            chunk: position for position, chunk in enumerate(bge_order)
        }
        missing_chunks = [
            chunk for chunk in baseline_order[:10] if chunk not in bge_rank
        ]
        if missing_chunks:
            blocked.append({
                "query_id": query_id,
                "reason": "BASELINE_TOP10_ITEMS_MISSING_FROM_STORED_BGE_ORDER",
                "chunk_ids": missing_chunks,
                "automatic_inference": False
            })
            continue

        baseline_rank = {
            chunk: position for position, chunk in enumerate(baseline_order)
        }
        candidate_top10 = sorted(
            baseline_order[:10],
            key=lambda chunk: (bge_rank[chunk], baseline_rank[chunk])
        )
        candidate_order = candidate_top10 + baseline_order[10:]

        need(
            set(candidate_top10) == set(baseline_order[:10])
            and candidate_order[10:] == baseline_order[10:]
            and len(candidate_order) == len(set(candidate_order))
            == len(baseline_order),
            "TOP10_MEMBERSHIP_OR_TAIL_CHANGED"
        )

        baseline_metrics = history_index[query_id]["before"]
        for metric in metric_names:
            need(
                abs(
                    baseline_metrics[metric]
                    - query_index[query_id]["baseline"][metric]
                ) <= 1e-9
                and abs(
                    baseline_metrics[metric]
                    - baseline_row["metrics"][metric]
                ) <= 1e-9,
                "BASELINE_STORED_METRIC_CONFLICT"
            )

        measured = measure_top10(candidate_order, set(gold_map[query_id]))
        need(
            abs(measured["recall_at_10"]
                - baseline_metrics["recall_at_10"]) <= 1e-9
            and measured["hit_at_10"] == baseline_metrics["hit_at_10"],
            "MEMBERSHIP_INVARIANT_METRIC_CONFLICT"
        )

        completed.append({
            "query_id": query_id,
            "baseline_top10": baseline_order[:10],
            "candidate_top10": candidate_top10,
            "baseline_tail_unchanged": True,
            "baseline_metrics_reused": baseline_metrics,
            "candidate_metrics": measured,
            "delta": {
                metric: measured[metric] - baseline_metrics[metric]
                for metric in metric_names
            }
        })

    complete = len(completed) == 24 and not blocked
    macro = None
    decision = "BLOCKED_INCOMPLETE_CACHE_COVERAGE_NO_EXPANSION"

    if complete:
        macro = {
            "baseline": {
                metric: math.fsum(
                    row["baseline_metrics_reused"][metric]
                    for row in completed
                ) / 24
                for metric in metric_names
            },
            "candidate": {
                metric: math.fsum(
                    row["candidate_metrics"][metric]
                    for row in completed
                ) / 24
                for metric in metric_names
            }
        }
        delta = {
            metric: macro["candidate"][metric] - macro["baseline"][metric]
            for metric in metric_names
        }
        macro["delta"] = delta
        decision = (
            "KEEP_AS_EXPLORATORY_CANDIDATE_NOT_ACCEPTED"
            if delta["ndcg_at_10"] > 1e-12
            else "REJECT_HYPOTHESIS_IN_THIS_COHORT"
        )

    result = {
        "schema_version": "NB22E_V3_C023_R09_V1",
        "status": (
            "R09_EXPLORATORY_EXPERIMENT_COMPLETE"
            if complete else "R09_EXPERIMENT_BLOCKED_BY_MISSING_STORED_RANKS"
        ),
        "roadmap_action": "R09",
        "plan_receipt": plan_output,
        "input_receipts": receipts,
        "planned_queries": 24,
        "completed_queries": len(completed),
        "blocked_queries": blocked,
        "per_query": completed,
        "macro": macro,
        "decision": decision,
        "ndcg_query_changes": {
            "improved": sum(
                row["delta"]["ndcg_at_10"] > 1e-12 for row in completed
            ),
            "worsened": sum(
                row["delta"]["ndcg_at_10"] < -1e-12 for row in completed
            ),
            "unchanged": sum(
                abs(row["delta"]["ndcg_at_10"]) <= 1e-12
                for row in completed
            )
        },
        "subtasks": {
            "R09.1": "DONE_ONE_EXPLICIT_EXPLORATORY_HYPOTHESIS_AND_PLAN",
            "R09.2": "NOT_APPLICABLE_NO_FEATURE_ABLATION_SELECTED",
            "R09.3": "NOT_APPLICABLE_NO_SHORTLIST_EXPANSION",
            "R09.4": "NOT_APPLICABLE_NO_ONNX_RUNTIME_USED",
            "R09.5": "DONE_ZERO_NEW_PAIRS_LOCAL_RANK_OPERATIONS_ONLY",
            "R09.6": (
                "DONE_SINGLE_EXPERIMENT_DECISION_RECORDED"
                if complete else "OPEN_CACHE_COVERAGE_BLOCKER_RECORDED"
            )
        },
        "all_applicable_R09_items_complete_in_exploratory_scope": complete,
        "elapsed_seconds": time.monotonic() - started,
        "new_api_calls": 0,
        "new_model_pairs": 0,
        "new_semantic_judgments": 0,
        "candidate_metrics_calculated": bool(completed),
        "previous_C073_C094_comparisons_recomputed": False,
        "formal_reference_qualified": False,
        "statistical_superiority_established": False,
        "baseline_replacement_authorized": False,
        "historical_gold_modified": False,
        "gate_changed": False,
        "automatic_expansion": False,
        "advance_to_next_roadmap_action": False,
        "next_step": "REVIEW_SINGLE_EXPERIMENT_RESULT_BEFORE_ANY_NEXT_ACTION"
    }

    output = save("C023.R09_top10_membership_experiment", result)

    print(json.dumps({
        "status": result["status"],
        "roadmap_action": "R09",
        "completed_queries": len(completed),
        "blocked_queries": blocked,
        "macro": macro,
        "ndcg_query_changes": result["ndcg_query_changes"],
        "decision": decision,
        "all_applicable_R09_items_complete_in_exploratory_scope": complete,
        "new_api_calls": 0,
        "new_model_pairs": 0,
        "baseline_replacement_authorized": False,
        "automatic_expansion": False,
        "advance_to_next_roadmap_action": False,
        "plan_output": plan_output,
        "write_status": "WRITTEN",
        "output": output
    }, ensure_ascii=False, indent=2))
    return result


V3_C023 = run_v3_c023()
````

````python
# C024 — R10: فریز baseline منتخب و حفظ نامزد اکتشافی
# بدون import، اجرای نوت‌بوک قبلی، API، مدل یا Metric تازه

def run_v3_c024():
    required = ("Path", "os", "json", "hashlib", "ast")
    missing = [name for name in required if name not in globals()]
    if missing:
        raise RuntimeError(
            "ایمپورت‌های زیر باید در سل اول موجود باشند: "
            + ", ".join(missing)
        )

    def need(condition, message):
        if not condition:
            raise ValueError(message)

    def sha(raw):
        return hashlib.sha256(raw).hexdigest()

    def canonical(value):
        return json.dumps(
            value, ensure_ascii=False, sort_keys=True,
            separators=(",", ":"), allow_nan=False
        ).encode("utf-8")

    def unique_object(items):
        result = {}
        for key, value in items:
            need(key not in result, "DUPLICATE_JSON_KEY: " + key)
            result[key] = value
        return result

    def reject_constant(value):
        raise ValueError("INVALID_JSON_CONSTANT: " + value)

    def parse(raw):
        return json.loads(
            raw.decode("utf-8"),
            object_pairs_hook=unique_object,
            parse_constant=reject_constant
        )

    root = Path(os.environ.get(
        "RAG_FINANCE_ROOT",
        "/Users/macbookpro/AMIR_DATA/00_WORKS/01_PROJECTS/"
        "RAG-projects/RAG_finance"
    )).expanduser().resolve()
    out_dir = root / "reports/phase2/quality_improvement/NB22E.lab.v3"
    receipts = {}

    def checked_path(relative):
        path = (root / relative).resolve()
        need(path.is_relative_to(root), "PATH_OUTSIDE_PROJECT")
        need(path.is_file(), "FILE_NOT_FOUND: " + relative)
        return path

    def read_bound(relative, digest):
        path = checked_path(relative)
        raw = path.read_bytes()
        need(sha(raw) == digest, "HASH_CONFLICT: " + relative)
        receipts[relative] = digest
        return parse(raw)

    def read_v3(prefix, digest):
        return read_bound(
            str((out_dir / (prefix + digest + ".json")).relative_to(root)),
            digest
        )

    def verify_stream(binding):
        relative = binding["path"]
        path = checked_path(relative)
        expected = binding["file_bytes_sha256"]
        fingerprint = hashlib.sha256()
        before = path.stat()
        with path.open("rb") as handle:
            for block in iter(lambda: handle.read(1024 * 1024), b""):
                fingerprint.update(block)
        after = path.stat()
        need(
            (before.st_size, before.st_mtime_ns, before.st_ino)
            == (after.st_size, after.st_mtime_ns, after.st_ino),
            "INPUT_CHANGED_DURING_READ: " + relative
        )
        need(
            fingerprint.hexdigest() == expected,
            "SOURCE_HASH_CONFLICT: " + relative
        )
        receipts[relative] = expected

    experiment = read_v3(
        "C023.R09_top10_membership_experiment.",
        "3dc2ad16a7475e401f6031ace0f92cb258f27f18bd48c3008b39e1583e080ab6"
    )
    experiment_plan = read_v3(
        "C023.R09_experiment_plan.",
        "e4ed586627038ad85b25807236d763401022d4a0928fe48a99a68a039e4f5384"
    )
    package = read_v3(
        "C013.R05_binding_inputs.",
        "4b10aaa2eab0e1110854fe0d2374be401ca6197f1aa7c4c9a902d1ff47ad820c"
    )

    need(
        experiment["all_applicable_R09_items_complete_in_exploratory_scope"]
        is True
        and experiment["completed_queries"] == 24
        and not experiment["blocked_queries"],
        "R09_NOT_COMPLETE"
    )
    need(
        experiment["decision"] == "KEEP_AS_EXPLORATORY_CANDIDATE_NOT_ACCEPTED"
        and experiment["baseline_replacement_authorized"] is False,
        "R09_SELECTION_STATE_CONFLICT"
    )
    need(
        experiment["plan_receipt"]["file_bytes_sha256"]
        == "e4ed586627038ad85b25807236d763401022d4a0928fe48a99a68a039e4f5384",
        "EXPERIMENT_PLAN_BINDING_CONFLICT"
    )
    need(
        experiment_plan["version"] == "BASELINE_TOP10_MEMBERSHIP_BGE_ORDER_V1"
        and experiment_plan["independent_holdout"] is False,
        "EXPERIMENT_SCOPE_CONFLICT"
    )

    def packaged_item(prefix=None, role=None):
        matches = [
            item for item in package["files"]
            if (
                prefix is not None
                and Path(item["path"]).name.startswith(prefix)
            ) or (
                role is not None and item["role"] == role
            )
        ]
        need(len(matches) == 1, "PACKAGE_ITEM_NOT_UNIQUE")
        item = matches[0]
        content = read_bound(item["path"], item["file_bytes_sha256"])
        need(content == item["content"], "PACKAGE_CONTENT_CONFLICT")
        return content, {
            "path": item["path"],
            "file_bytes_sha256": item["file_bytes_sha256"]
        }

    source_plan, source_plan_receipt = packaged_item(
        prefix="C063.full_pool24_plan."
    )
    metric_contract, metric_contract_receipt = packaged_item(
        role="C025_metric_contract"
    )

    producer_report_sha = (
        "4b1626f9669aa7bc8ffa0ada5ddffe8d"
        "7350aa8dcf5db9e956bd8c67f86c82de"
    )
    producer_report_path = (
        "reports/phase2/quality_improvement/NB22D.v1/"
        "A05.final_report." + producer_report_sha + ".json"
    )
    producer_report = read_bound(
        producer_report_path, producer_report_sha
    )

    baseline_binding = source_plan["source_bindings"]["baseline"]
    need(
        producer_report["retained_rankings"]["path"]
        == baseline_binding["path"]
        and producer_report["retained_rankings"]["file_bytes_sha256"]
        == baseline_binding["file_bytes_sha256"],
        "BASELINE_PRODUCER_RECEIPT_CONFLICT"
    )
    need(
        producer_report["release_acceptance_granted"] is False,
        "PRODUCER_ACCEPTANCE_SCOPE_CONFLICT"
    )

    # فقط تطبیق بایت داده‌ها؛ هیچ رتبه یا Gold دوباره ساخته نمی‌شود.
    for name in ("baseline", "pool", "chunks", "golden"):
        verify_stream(source_plan["source_bindings"][name])

    # اتصال کد واقعی Producer به گزارش تاریخی، بدون اجرای کد.
    notebook_relative = (
        "notebooks/phase2/22d_full_corpus_retrieval_expansion.ipynb"
    )
    notebook_path = checked_path(notebook_relative)
    notebook_raw = notebook_path.read_bytes()
    notebook = parse(notebook_raw)
    code_cells = []
    for cell in notebook["cells"]:
        if cell["cell_type"] != "code":
            continue
        source = cell["source"]
        code_cells.append(
            source if isinstance(source, str) else "".join(source)
        )

    code_sha = sha(canonical(code_cells))
    need(
        code_sha == producer_report["producer_code_sha256"],
        "PRODUCER_CODE_DIFFERS_FROM_HISTORICAL_REPORT"
    )
    need(
        notebook.get("metadata", {}).get("nb22d", {})
        .get("code_source_sha256") == code_sha,
        "PRODUCER_NOTEBOOK_METADATA_CONFLICT"
    )
    receipts[notebook_relative] = sha(notebook_raw)

    # خواندن CONFIG از AST؛ هیچ eval یا اجرای کد Producer انجام نمی‌شود.
    configurations = []
    for source in code_cells:
        tree = ast.parse(source)
        for node in tree.body:
            if isinstance(node, ast.Assign) and any(
                isinstance(target, ast.Name) and target.id == "CONFIG"
                for target in node.targets
            ):
                configurations.append(ast.literal_eval(node.value))

    need(len(configurations) == 1, "PRODUCER_CONFIG_NOT_UNIQUE")
    producer_config = configurations[0]
    need(isinstance(producer_config, dict), "PRODUCER_CONFIG_NOT_OBJECT")

    need(
        producer_config["thresholds"]["recall_at_10"]
        == metric_contract["metric_policy"]["recall_at_k"]
        ["absolute_gate"]["value"]
        and producer_config["thresholds"]["ndcg_at_10"]
        == metric_contract["metric_policy"]["ndcg_at_k_binary"]
        ["absolute_gate"]["value"],
        "FROZEN_THRESHOLD_CONFLICT"
    )

    result = {
        "schema_version": "NB22E_V3_C024_R10_V1",
        "status": "R10_BASELINE_AND_EVALUATION_EVIDENCE_FROZEN",
        "roadmap_action": "R10",
        "input_receipts": receipts,
        "selected_method": {
            "role": "RETAINED_BASELINE_NOT_NEWLY_ACCEPTED_RELEASE",
            "variant": producer_report["retained_ranking"],
            "ranking_artifact": baseline_binding,
            "selection_report": {
                "path": producer_report_path,
                "file_bytes_sha256": producer_report_sha
            },
            "producer_notebook": notebook_relative,
            "producer_code_sha256": code_sha,
            "producer_code_matches_historical_report": True,
            "producer_code_snapshot": code_cells,
            "configuration": producer_config,
            "configuration_canonical_sha256": sha(canonical(producer_config)),
            "runtime_executed_or_health_checked": False
        },
        "evaluation_lock": {
            "source_plan": source_plan_receipt,
            "data_bindings": source_plan["source_bindings"],
            "historical_gold_mapping_sha256":
                source_plan["historical_gold_sha256"],
            "question_texts_sha256": source_plan["question_texts_sha256"],
            "metric_contract": metric_contract_receipt,
            "retrieval_metric_policies": {
                key: metric_contract["metric_policy"][key]
                for key in ("recall_at_k", "ndcg_at_k_binary", "hit_at_k")
            },
            "historical_gold_modified": False,
            "thresholds_changed": False
        },
        "preserved_exploratory_candidate": {
            "version": experiment_plan["version"],
            "plan_receipt": experiment["plan_receipt"],
            "result_receipt": {
                "path": (
                    "reports/phase2/quality_improvement/NB22E.lab.v3/"
                    "C023.R09_top10_membership_experiment."
                    "3dc2ad16a7475e401f6031ace0f92cb258f27f18bd48c3008b39e1583e080ab6.json"
                ),
                "file_bytes_sha256": (
                    "3dc2ad16a7475e401f6031ace0f92cb258f27f18bd48c3008b39e1583e080ab6"
                )
            },
            "metrics_reused": experiment["macro"],
            "decision": experiment["decision"],
            "baseline_replacement_authorized": False,
            "requires_further_evidence_before_selection": True
        },
        "exposure_record": {
            "producer_scope_as_recorded": producer_report["scope"],
            "current_experiment_population":
                "PREVIOUSLY_EXPOSED_DEVELOPMENT24",
            "current_experiment_query_ids": experiment_plan["cohort"],
            "current_experiment_is_independent_holdout": False,
            "historical_test_previously_seen": True,
            "new_test_execution_performed": False,
            "historical_test_used_for_current_experiment_selection": False,
            "complete_projectwide_exposure_history_established": False,
            "unseen_prior_exposures_inferred": False
        },
        "subtasks": {
            "R10.1": {
                "status": "DONE_RETAINED_BASELINE_SELECTED_NEW_CANDIDATE_SEPARATE",
                "basis_fa": (
                    "baseline منتخب قبلی حفظ شد؛ "
                    "نامزد ۲۴سؤالی جایگزین آن نشده است."
                )
            },
            "R10.2": {
                "status": "DONE_EVIDENCE_METHOD_CODE_CONFIG_AND_DATA_LOCKED",
                "basis_fa": (
                    "رسید رتبه‌ها، گزارش انتخاب، هش کد تاریخی، "
                    "تنظیمات، داده‌ها و قرارداد معیار متصل شدند."
                )
            },
            "R10.3": {
                "status": "DONE_KNOWN_EXPOSURES_AND_LIMITS_RECORDED",
                "basis_fa": (
                    "دیده‌شدن Development و Test تاریخی ثبت شد؛ "
                    "استقلال یا تاریخچهٔ کامل مواجهه ادعا نشده است."
                )
            },
            "R10.4": {
                "status": "DONE_NO_NEW_TEST_OR_INDEPENDENCE_CLAIM",
                "basis_fa": (
                    "Test تازه اجرا نشد؛ ارزیابی مستقل آینده "
                    "به طرح و دامنهٔ جدا نیاز دارد."
                )
            }
        },
        "all_R10_items_complete_in_evidence_freeze_scope": True,
        "live_product_runtime_frozen_or_verified": False,
        "formal_reference_qualified": False,
        "independent_final_evaluation_complete": False,
        "release_accepted": False,
        "gate_changed": False,
        "new_api_calls": 0,
        "new_model_pairs": 0,
        "new_semantic_judgments": 0,
        "metrics_recomputed": False,
        "advance_to_next_roadmap_action": False,
        "next_step": "REPORT_R10_FREEZE_BEFORE_STARTING_R11"
    }

    raw = (
        json.dumps(
            result, ensure_ascii=False, sort_keys=True,
            indent=2, allow_nan=False
        ) + "\n"
    ).encode("utf-8")
    digest = sha(raw)
    target = out_dir / (
        "C024.R10_selected_evidence_freeze." + digest + ".json"
    )
    if target.exists():
        need(target.read_bytes() == raw, "OUTPUT_CONTENT_CONFLICT")
    else:
        with target.open("xb") as handle:
            handle.write(raw)

    saved = target.read_bytes()
    need(sha(saved) == digest, "OUTPUT_HASH_CONFLICT")
    need(parse(saved) == result, "OUTPUT_RELOAD_CONFLICT")

    print(json.dumps({
        "status": result["status"],
        "roadmap_action": "R10",
        "selected_variant": producer_report["retained_ranking"],
        "producer_code_matches_historical_report": True,
        "exploratory_candidate_preserved_separately": True,
        "all_R10_items_complete_in_evidence_freeze_scope": True,
        "live_product_runtime_frozen_or_verified": False,
        "formal_reference_qualified": False,
        "release_accepted": False,
        "new_api_calls": 0,
        "new_model_pairs": 0,
        "metrics_recomputed": False,
        "advance_to_next_roadmap_action": False,
        "write_status": "WRITTEN",
        "output": {
            "path": str(target.relative_to(root)),
            "file_bytes_sha256": digest,
            "reload_validation": "PASS"
        }
    }, ensure_ascii=False, indent=2))
    return result


V3_C024 = run_v3_c024()
````

````python
# C025 — R11: مصرف Gate کامل و اتصال فقط ۱۶۰ رأی بازیابی‌شده
# بدون import، API، داوری تازه، اجرای مدل یا محاسبهٔ Metric

def run_v3_c025():
    required = ("Path", "os", "json", "hashlib")
    missing = [name for name in required if name not in globals()]
    if missing:
        raise RuntimeError(
            "ابتدا سل imports را اجرا کنید: " + ", ".join(missing)
        )

    def need(ok, message):
        if not ok:
            raise ValueError(message)

    def sha(raw):
        return hashlib.sha256(raw).hexdigest()

    def unique_object(items):
        result = {}
        for key, value in items:
            need(key not in result, "DUPLICATE_JSON_KEY: " + key)
            result[key] = value
        return result

    def reject_constant(value):
        raise ValueError("INVALID_JSON_CONSTANT: " + value)

    def parse(raw):
        return json.loads(
            raw.decode("utf-8"),
            object_pairs_hook=unique_object,
            parse_constant=reject_constant
        )

    root = Path(os.environ.get(
        "RAG_FINANCE_ROOT",
        "/Users/macbookpro/AMIR_DATA/00_WORKS/01_PROJECTS/"
        "RAG-projects/RAG_finance"
    )).expanduser().resolve()
    v2 = root / "reports/phase2/quality_improvement/NB22E.lab.v2"
    v3 = root / "reports/phase2/quality_improvement/NB22E.lab.v3"
    receipts = {}

    def read_bound(relative, digest, jsonl=False):
        path = (root / relative).resolve()
        need(path.is_relative_to(root), "PATH_OUTSIDE_PROJECT")
        need(path.is_file(), "FILE_NOT_FOUND: " + relative)
        raw = path.read_bytes()
        need(sha(raw) == digest, "HASH_CONFLICT: " + relative)
        receipts[relative] = digest
        return (
            [parse(line) for line in raw.splitlines() if line.strip()]
            if jsonl else parse(raw)
        )

    def read_v3(prefix, digest):
        return read_bound(
            str((v3 / (prefix + digest + ".json")).relative_to(root)),
            digest
        )

    frozen = read_v3(
        "C024.R10_selected_evidence_freeze.",
        "b807237f302b824c5ad943b5e3adb68a79ebd5c29c51c30edc488b88deec82e8"
    )
    recovery = read_v3(
        "C021.R08_recovery_disposition.",
        "72b075dd2ec734a8182e3d1dbe012f32494f5b78ba994ac543108be704f78cd1"
    )

    ledger_sha = (
        "565cd36340044a6c879045db52d6ac8df"
        "74eea8e1e3823556202b4541a5d1abf"
    )
    ledger = read_bound(
        str((v2 / (
            "C112.review_work_queue." + ledger_sha + ".json"
        )).relative_to(root)),
        ledger_sha
    )

    need(
        frozen["all_R10_items_complete_in_evidence_freeze_scope"] is True,
        "R10_FREEZE_NOT_COMPLETE"
    )
    need(
        recovery["R08_recovery_assessment_complete_in_declared_scope"] is True,
        "R08_RECOVERY_ASSESSMENT_NOT_COMPLETE"
    )

    required_sections = (
        "R11_metric_gap_matrix_C100",
        "R11_evidence_reconciliation_C102",
        "exact_test_unit_audit_C104",
        "A07_exact_unit_binding_C106",
        "producer_exact_requirement_binding_C108"
    )
    for section in required_sections:
        need(
            section in ledger,
            "MISSING_EXISTING_LEDGER_SECTION: " + section
        )

    gates = ledger["R11_metric_gap_matrix_C100"]
    reconciliation = ledger["R11_evidence_reconciliation_C102"]
    audit = ledger["exact_test_unit_audit_C104"]
    old_binding = ledger["A07_exact_unit_binding_C106"]

    # کنترل صحیح: جدول کامل و زیرمجموعهٔ BLOCKED جدا هستند.
    full_matrix = gates["metric_gap_matrix"]
    need(isinstance(full_matrix, list), "GATE_MATRIX_NOT_LIST")
    need(
        len({row["metric"] for row in full_matrix}) == len(full_matrix),
        "DUPLICATE_GATE_METRIC"
    )

    historical_counts = {"PASS": 0, "FAIL": 0, "BLOCKED": 0}
    for row in full_matrix:
        status = row["historical_status"]
        need(status in historical_counts, "UNKNOWN_HISTORICAL_GATE_STATUS")
        historical_counts[status] += 1

    need(
        historical_counts == gates["historical_metric_status_counts"],
        "GATE_COUNTS_DISAGREE_WITH_RECORDED_SUMMARY"
    )
    need(
        historical_counts == {"PASS": 2, "FAIL": 2, "BLOCKED": 12}
        and len(full_matrix) == 16,
        "HISTORICAL_GATE_POPULATION_CONFLICT"
    )

    blocked_matrix = [
        row for row in full_matrix
        if row["historical_status"] == "BLOCKED"
    ]
    need(
        len(blocked_matrix) == 12
        and len({row["metric"] for row in blocked_matrix}) == 12,
        "BLOCKED_SUBSET_CONFLICT"
    )

    # اتصال‌های سالم قبلی فقط مصرف می‌شوند.
    old_matched = [
        row for row in old_binding["case_results"]
        if row["binding_status"] == "EXACT_RESPONSE_DOCUMENT_MATCH"
    ]
    old_ids = {row["candidate_id"] for row in old_matched}
    need(
        len(old_matched) == len(old_ids) == 887,
        "OLD_BINDING_COUNT_CONFLICT"
    )

    def audit_table(name, jsonl):
        binding = audit["table_summaries"][name]
        return read_bound(
            binding["path"], binding["file_bytes_sha256"], jsonl
        )

    cases = audit_table("cases", False)
    claims = audit_table("claims", True)
    links = audit_table("links", True)

    case_index = {row["candidate_id"]: row for row in cases}
    claim_index = {
        (row["candidate_id"], row["claim_index"]): row for row in claims
    }
    link_index = {
        (row["candidate_id"], row["claim_index"], row["evidence_id"]): row
        for row in links
    }

    need(len(case_index) == len(cases) == 1149, "CASE_POPULATION_CONFLICT")
    need(len(claim_index) == len(claims), "DUPLICATE_CLAIM_KEY")
    need(len(link_index) == len(links), "DUPLICATE_LINK_KEY")
    need(old_ids.issubset(case_index), "OLD_CASE_OUTSIDE_POPULATION")

    recovered_rows = [
        row for row in recovery["outcomes"]
        if row["status"] == "RECOVERED_LOCALLY"
    ]
    recovered_ids = {row["candidate_id"] for row in recovered_rows}
    need(
        len(recovered_rows) == len(recovered_ids) == 160
        and not (old_ids & recovered_ids),
        "RECOVERED_POPULATION_CONFLICT"
    )

    newly_bound = []
    bound_links = set()

    for row in recovered_rows:
        candidate_id = row["candidate_id"]
        need(candidate_id in case_index, "RECOVERED_CASE_NOT_IN_C068")
        case = case_index[candidate_id]
        review = row["validated_review"]["result"]

        need(
            review["candidate_id"] == candidate_id
            and review["response_document_sha256"]
            == case["response_document_sha256"],
            candidate_id + ": RESPONSE_VERSION_CONFLICT"
        )
        need(
            case["response_and_support_version_verified"] is True,
            candidate_id + ": HISTORICAL_RESPONSE_BINDING_NOT_VERIFIED"
        )

        expected_links = {
            key for key in link_index if key[0] == candidate_id
        }
        reviewed_links = set()

        for item in review["citation_links"]:
            key = (
                candidate_id, item["claim_index"], item["evidence_id"]
            )
            need(key not in reviewed_links, "DUPLICATE_RECOVERED_LINK")
            need(key in link_index, candidate_id + ": LINK_NOT_IN_C068")
            need(
                (candidate_id, item["claim_index"]) in claim_index,
                candidate_id + ": CLAIM_NOT_IN_C068"
            )
            reviewed_links.add(key)

        need(
            reviewed_links == expected_links,
            candidate_id + ": DISTINCT_LINK_SET_CONFLICT"
        )
        bound_links.update(reviewed_links)

        newly_bound.append({
            "candidate_id": candidate_id,
            "response_document_sha256": case["response_document_sha256"],
            "binding_status": "EXACT_HISTORICAL_RESPONSE_AND_LINK_SET_MATCH",
            "source_attempt_directory": row["source_directory"],
            "link_count": len(reviewed_links),
            "requirement_review_count": len(review["requirement_coverage"]),
            "additional_factual_unit_count":
                len(review["additional_factual_units"]),
            "attribution_assertion_count":
                len(review["attribution_assertions"]),
            "reference_requirement_qualification":
                review["reference_requirement_qualification"],
            "administrative_reason_generated":
                row["administrative_reason_generated"],
            "review": review,
            "independent_semantic_verification": False,
            "formal_reference_qualified": False,
            "transferred_to_selected_retrieval_response": False
        })

    all_locally_bound_ids = old_ids | recovered_ids
    remaining_ids = sorted(set(case_index) - all_locally_bound_ids)

    need(
        len(all_locally_bound_ids) == 1047
        and len(remaining_ids) == 102,
        "FINAL_BINDING_ACCOUNTING_CONFLICT"
    )

    recovery_remaining = {
        row["candidate_id"] for row in recovery["outcomes"]
        if row["status"] != "RECOVERED_LOCALLY"
    }
    need(
        set(remaining_ids) == recovery_remaining,
        "R08_R11_RESIDUAL_IDENTITY_CONFLICT"
    )

    semantic_metrics = {
        "answer_completeness", "attribution_correctness",
        "citation_correctness", "citation_locator_accuracy",
        "claim_citation_coverage", "groundedness"
    }

    updated_gaps = []
    for historical in blocked_matrix:
        metric = historical["metric"]
        updated_gaps.append({
            "metric": metric,
            "historical_gap_record_preserved": historical,
            "existing_reconciliation_preserved_in_source": True,
            "new_evidence_available": (
                "160_RECOVERED_HISTORICAL_RESPONSE_REVIEWS_BOUND"
                if metric in semantic_metrics
                else "NO_NEW_METRIC_EVIDENCE_FROM_THIS_JOIN"
            ),
            "new_evidence_is_selected_method_evaluation": False,
            "historical_gate_overwritten": False,
            "current_selected_method_gate_status": "NOT_EVALUATED",
            "remaining_task": (
                "CHECK_CURRENT_METRIC_UNIT_REFERENCE_DENOMINATOR_AND_VERSION"
                if metric in semantic_metrics
                else historical["remaining_work"]
            )
        })

    result = {
        "schema_version": "NB22E_V3_C025_R11_V2",
        "status": "R11_EXISTING_GATES_CONSUMED_RECOVERED_REVIEWS_BOUND",
        "roadmap_action": "R11",
        "input_receipts": receipts,
        "selected_variant": frozen["selected_method"]["variant"],
        "historical_gate": {
            "source": gates["historical_gate_source"],
            "overall_status": gates["historical_overall_gate"],
            "metric_status_counts": historical_counts,
            "full_metric_count": len(full_matrix),
            "blocked_metric_count": len(blocked_matrix),
            "full_metric_matrix_preserved": full_matrix,
            "overwritten": False
        },
        "existing_reconciliation": reconciliation,
        "updated_blocked_metric_gap_matrix": updated_gaps,
        "old_successful_bindings_reused": len(old_ids),
        "newly_recovered_case_bindings": newly_bound,
        "newly_bound_distinct_link_count": len(bound_links),
        "locally_valid_and_historically_bound_cases":
            len(all_locally_bound_ids),
        "remaining_historical_case_ids": remaining_ids,
        "remaining_historical_case_count": len(remaining_ids),
        "remaining_states_from_R08": {
            "still_held": recovery["still_held_count"],
            "received_http_failure": recovery["received_http_failure_count"]
        },
        "regression_specifications_preserved":
            gates["regression_specification_summary"],
        "new_regression_tests_executed": 0,
        "integrity_checks_are_regression_suite_execution": False,
        "provenance_boundaries": {
            "judgments_scope": "HISTORICAL_TEST_RESPONSES",
            "selected_method": frozen["selected_method"]["variant"],
            "selected_method_response_quality_established": False,
            "reviewer_is_product_answer_generator": False,
            "API_cost_is_product_runtime_cost": False,
            "structural_validation_is_semantic_acceptance": False
        },
        "subtasks": {
            "R11.1":
                "DONE_EXISTING_TWELVE_BLOCKED_METRIC_RECORDS_CONSUMED",
            "R11.2":
                "PARTIAL_RECOVERED_HISTORICAL_UNITS_BOUND_METRIC_READINESS_OPEN",
            "R11.3":
                "OPEN_REQUIRED_EXECUTABLE_REGRESSION_RESULTS_NOT_ESTABLISHED",
            "R11.4":
                "DONE_HISTORICAL_RESPONSE_REVIEWER_AND_PRODUCT_SCOPE_SEPARATED",
            "R11.5":
                "OPEN_CURRENT_GATE_NOT_EVALUATED"
        },
        "all_R11_items_complete": False,
        "new_api_calls": 0,
        "new_model_pairs": 0,
        "new_semantic_judgments": 0,
        "metrics_calculated": False,
        "historical_files_modified": False,
        "gate_pass_granted": False,
        "advance_to_next_roadmap_action": False,
        "next_step": (
            "DETERMINE_METRIC_READINESS_AND_REQUIRED_REGRESSION_"
            "FROM_BOUND_EVIDENCE"
        )
    }

    raw = (
        json.dumps(
            result, ensure_ascii=False, sort_keys=True,
            indent=2, allow_nan=False
        ) + "\n"
    ).encode("utf-8")
    digest = sha(raw)
    target = v3 / (
        "C025.R11_recovered_evidence_binding." + digest + ".json"
    )
    if target.exists():
        need(target.read_bytes() == raw, "OUTPUT_CONTENT_CONFLICT")
    else:
        with target.open("xb") as handle:
            handle.write(raw)

    saved = target.read_bytes()
    need(sha(saved) == digest, "OUTPUT_HASH_CONFLICT")
    need(parse(saved) == result, "OUTPUT_RELOAD_CONFLICT")

    print(json.dumps({
        "status": result["status"],
        "roadmap_action": "R11",
        "selected_variant": result["selected_variant"],
        "historical_full_metric_count": len(full_matrix),
        "historical_blocked_metric_count": len(blocked_matrix),
        "historical_metric_status_counts": historical_counts,
        "old_successful_bindings_reused": len(old_ids),
        "newly_recovered_cases_bound": len(newly_bound),
        "newly_bound_distinct_link_count": len(bound_links),
        "locally_valid_and_historically_bound_cases":
            len(all_locally_bound_ids),
        "remaining_historical_case_count": len(remaining_ids),
        "subtasks": result["subtasks"],
        "new_regression_tests_executed": 0,
        "all_R11_items_complete": False,
        "metrics_calculated": False,
        "new_api_calls": 0,
        "gate_pass_granted": False,
        "advance_to_next_roadmap_action": False,
        "write_status": "WRITTEN",
        "output": {
            "path": str(target.relative_to(root)),
            "file_bytes_sha256": digest,
            "reload_validation": "PASS"
        }
    }, ensure_ascii=False, indent=2))
    return result


V3_C025 = run_v3_c025()
````

### ادامهٔ کار یازده — کفایت واحدها و ورودی واقعی Regression

این بسته:
- ۸۸۷ رأی قبلی و ۱۶۰ رأی بازیابی‌شده را بدون بازداوری مصرف می‌کند.
- نسخهٔ پاسخ و پوشش پیوندهای هر پرونده را کنترل می‌کند.
- متن و ترتیب نیازمندی‌ها را با ورودی تاریخی تطبیق می‌دهد.
- نیازمندی غایب یا null را «بدون نیازمندی» تفسیر نمی‌کند.
- شمار وضعیت‌های پوشش، پشتیبانی، محل و انتساب را جدا ثبت می‌کند.
- مشخصات اصلی Regression و شواهد قبلی مرتبط را مصرف می‌کند.

تطبیق نیازمندی، صلاحیت معنایی Rubric نیست.
محلِ تأییدشده توسط مدل، تأیید کامل Locator منبع نیست.
خالی‌بودن فهرست انتساب، کامل‌بودن استخراج را اثبات نمی‌کند.

شمار پرونده، ادعا، پیوند، نیازمندی و تست جدا گزارش می‌شود.
هیچ مشخصات تستی بدون اجرای واقعی PASS نام نمی‌گیرد.

خروجی، یک بستهٔ واحد برای تعیین سنجه‌های قابل‌مصرف
و تست‌های واقعاً قابل‌اجراست؛ بستهٔ جدیدِ داوری عمومی نیست.

API، تولید پاسخ و داوری تازه صفر است.
کار یازده هنوز باز می‌ماند.

````python
# C026 — R11: کنترل کفایت واحدها و خواندن مشخصات واقعی Regression
# بدون import، API، تولید پاسخ یا داوری تازه

def run_v3_c026():
    missing = [
        name for name in ("Path", "os", "json", "hashlib")
        if name not in globals()
    ]
    if missing:
        raise RuntimeError(
            "ابتدا سل imports را اجرا کنید: " + ", ".join(missing)
        )

    def need(ok, message):
        if not ok:
            raise ValueError(message)

    def sha(raw):
        return hashlib.sha256(raw).hexdigest()

    def unique_object(items):
        result = {}
        for key, value in items:
            need(key not in result, "DUPLICATE_JSON_KEY: " + key)
            result[key] = value
        return result

    def reject_constant(value):
        raise ValueError("INVALID_JSON_CONSTANT: " + value)

    def parse(raw):
        return json.loads(
            raw.decode("utf-8"),
            object_pairs_hook=unique_object,
            parse_constant=reject_constant
        )

    root = Path(os.environ.get(
        "RAG_FINANCE_ROOT",
        "/Users/macbookpro/AMIR_DATA/00_WORKS/01_PROJECTS/"
        "RAG-projects/RAG_finance"
    )).expanduser().resolve()
    v2 = root / "reports/phase2/quality_improvement/NB22E.lab.v2"
    v3 = root / "reports/phase2/quality_improvement/NB22E.lab.v3"
    receipts = {}

    def read_bound(relative, digest, jsonl=False):
        path = (root / relative).resolve()
        need(path.is_relative_to(root), "PATH_OUTSIDE_PROJECT")
        need(path.is_file(), "FILE_NOT_FOUND: " + relative)
        raw = path.read_bytes()
        need(sha(raw) == digest, "HASH_CONFLICT: " + relative)
        receipts[relative] = digest
        return (
            [parse(line) for line in raw.splitlines() if line.strip()]
            if jsonl else parse(raw)
        )

    def read_v3(prefix, digest):
        return read_bound(
            str((v3 / (prefix + digest + ".json")).relative_to(root)),
            digest
        )

    previous = read_v3(
        "C025.R11_recovered_evidence_binding.",
        "2fdd2e89e91c54c59076fca8b63760a3bf8e0c6546e13740a11733b9b670f55f"
    )
    original = read_v3(
        "C019.R08_existing_inputs.",
        "a4b58b0ab44a206c959fcfe0c6f06e640ac92af207509be7a95cfbc292029ced"
    )

    ledger_sha = (
        "565cd36340044a6c879045db52d6ac8df"
        "74eea8e1e3823556202b4541a5d1abf"
    )
    ledger = read_bound(
        str((v2 / (
            "C112.review_work_queue." + ledger_sha + ".json"
        )).relative_to(root)),
        ledger_sha
    )

    need(
        previous["locally_valid_and_historically_bound_cases"] == 1047,
        "C025_POPULATION_CONFLICT"
    )

    packet_binding = ledger["exact_test_unit_audit_C104"][
        "existing_residual_packets"
    ][0]
    packets_list = read_bound(
        packet_binding["path"],
        packet_binding["file_bytes_sha256"],
        jsonl=True
    )
    packets = {row["candidate_id"]: row for row in packets_list}
    need(len(packets) == len(packets_list) == 1149, "PACKET_IDENTITY_CONFLICT")

    # مصرف نتایج قبلی؛ Validator یا مدل دوباره اجرا نمی‌شود.
    reviews = {}
    provenance = {}
    for row in original["A07_results_preserved"]:
        if row["status"] != "LOCALLY_VALID_MODEL_GAP_REVIEW":
            continue
        candidate_id = row["candidate_id"]
        need(candidate_id not in reviews, "DUPLICATE_OLD_REVIEW")
        reviews[candidate_id] = row["result"]
        provenance[candidate_id] = "EXISTING_A07_VALID_REVIEW"

    need(len(reviews) == 887, "OLD_REVIEW_COUNT_CONFLICT")

    for row in previous["newly_recovered_case_bindings"]:
        candidate_id = row["candidate_id"]
        need(candidate_id not in reviews, "RECOVERED_REVIEW_OVERLAP")
        reviews[candidate_id] = row["review"]
        provenance[candidate_id] = "C021_RECOVERED_C025_BOUND"

    need(len(reviews) == 1047, "MERGED_REVIEW_COUNT_CONFLICT")
    need(set(reviews).issubset(packets), "REVIEW_OUTSIDE_PACKET_POPULATION")

    def count(container, value):
        key = "null" if value is None else str(value)
        container[key] = container.get(key, 0) + 1

    totals = {
        "response_version_matches": 0,
        "exact_link_sets": 0,
        "requirement_binding_states": {},
        "requirement_assessment_counts": {},
        "link_support_counts": {},
        "link_intended_location_counts": {},
        "additional_factual_assessment_counts": {},
        "attribution_role_counts": {},
        "attribution_assessment_counts": {},
        "factual_extraction_review_counts": {},
        "reference_qualification_counts": {}
    }
    per_case = []

    for candidate_id in sorted(reviews):
        review = reviews[candidate_id]
        packet = packets[candidate_id]

        need(
            review["candidate_id"] == candidate_id
            and review["response_document_sha256"]
            == packet["response_document_sha256"],
            candidate_id + ": RESPONSE_VERSION_CONFLICT"
        )
        totals["response_version_matches"] += 1

        response = packet["decoded_response"]
        expected_links = {
            (index, evidence_id)
            for index, claim in enumerate(response["claims"])
            for evidence_id in claim["evidence_ids"]
        }
        actual_links = [
            (row["claim_index"], row["evidence_id"])
            for row in review["citation_links"]
        ]
        need(
            len(actual_links) == len(set(actual_links))
            and set(actual_links) == expected_links,
            candidate_id + ": LINK_SET_CONFLICT"
        )
        totals["exact_link_sets"] += 1

        requirements = packet.get("recorded_requirements_from_action26")
        indices = [
            row["requirement_index"]
            for row in review["requirement_coverage"]
        ]

        if requirements is None:
            requirement_state = "ORIGINAL_RUBRIC_NULL_OR_ABSENT"
        elif isinstance(requirements, list) and all(
            isinstance(value, str) for value in requirements
        ):
            need(
                set(indices) == set(range(len(requirements)))
                and len(indices) == len(set(indices)),
                candidate_id + ": REQUIREMENT_INDEX_CONFLICT"
            )
            requirement_state = (
                "EXPLICIT_EMPTY_RECORDED_LIST"
                if not requirements
                else "RECORDED_REQUIREMENT_LIST_AND_INDICES_BOUND"
            )
        else:
            requirement_state = "OTHER_RUBRIC_SCHEMA_NOT_COERCED"

        count(totals["requirement_binding_states"], requirement_state)
        count(
            totals["factual_extraction_review_counts"],
            review["factual_extraction_review"]
        )
        count(
            totals["reference_qualification_counts"],
            review["reference_requirement_qualification"]
        )

        for row in review["requirement_coverage"]:
            count(totals["requirement_assessment_counts"], row["assessment"])
        for row in review["citation_links"]:
            count(totals["link_support_counts"], row["support"])
            count(
                totals["link_intended_location_counts"],
                row["intended_location"]
            )
        for row in review["additional_factual_units"]:
            count(
                totals["additional_factual_assessment_counts"],
                row["assessment"]
            )
        for row in review["attribution_assertions"]:
            count(totals["attribution_role_counts"], row["role"])
            count(
                totals["attribution_assessment_counts"],
                row["assessment"]
            )

        per_case.append({
            "candidate_id": candidate_id,
            "provenance": provenance[candidate_id],
            "response_document_sha256": packet["response_document_sha256"],
            "recorded_requirement_state": requirement_state,
            "recorded_requirements": requirements,
            "reference_requirement_qualification":
                review["reference_requirement_qualification"],
            "factual_extraction_review": review["factual_extraction_review"],
            "review": review,
            "original_essential_requirement_qualification_granted": False,
            "bound_to_selected_method_generated_response": False
        })

    # خواندن همان فایل مشخصات Regression که C100 قبلاً معرفی کرده است.
    gates = ledger["R11_metric_gap_matrix_C100"]
    regression_binding = gates["input_receipts"]["regression_scenarios322"]
    regression_digest = (
        regression_binding.get("file_bytes_sha256")
        or regression_binding.get("expected_file_bytes_sha256")
    )
    need(
        isinstance(regression_digest, str) and len(regression_digest) == 64,
        "REGRESSION_RECEIPT_HASH_MISSING"
    )
    regression_specs = read_bound(
        regression_binding["path"], regression_digest
    )
    need(
        isinstance(regression_specs, list) and len(regression_specs) == 322,
        "REGRESSION_SPECIFICATION_POPULATION_CONFLICT"
    )

    # فقط گزارش‌های قبلاً خوانده‌شده و مرتبط در زنجیرهٔ موجود.
    lineage = ledger["R11_bounded_evidence_lineage_C101"]
    relevant_fragments = (
        "C018.report.", "C040.report.",
        "C046.report.", "C047.report.", "C048.report.",
        "C069.metric_table.", "C455.handoff."
    )
    evidence_documents = [
        document for document in lineage["documents"]
        if "stored_content" in document
        and any(
            fragment in document["path"]
            for fragment in relevant_fragments
        )
    ]

    result = {
        "schema_version": "NB22E_V3_C026_R11_V1",
        "status": "R11_UNIT_READINESS_AUDITED_REGRESSION_SPECIFICATIONS_READ",
        "roadmap_action": "R11",
        "input_receipts": receipts,
        "historical_population": 1149,
        "reviewed_and_bound_population": len(per_case),
        "remaining_population": 1149 - len(per_case),
        "unit_accounting": totals,
        "per_case_unit_reviews": per_case,
        "unreviewed_case_ids": sorted(set(packets) - set(reviews)),
        "existing_metric_reconciliation":
            previous["existing_reconciliation"],
        "historical_gate_preserved": previous["historical_gate"],
        "regression_specifications": regression_specs,
        "regression_specification_count": len(regression_specs),
        "relevant_existing_evidence_documents": evidence_documents,
        "frozen_selected_method": previous["selected_variant"],
        "selected_method_answer_quality_established": False,
        "regression_tests_executed": 0,
        "specifications_are_executed_tests": False,
        "rubric_qualification_granted": False,
        "new_api_calls": 0,
        "new_model_pairs": 0,
        "new_semantic_judgments": 0,
        "metrics_calculated": False,
        "gate_pass_granted": False,
        "all_R11_items_complete": False,
        "advance_to_next_roadmap_action": False,
        "next_step": (
            "REVIEW_ACTUAL_REGRESSION_SPECIFICATIONS_AND_METRIC_EVIDENCE_"
            "TO_SELECT_ONLY_EXECUTABLE_REQUIRED_CHECKS"
        ),
        "user_action_fa": (
            "فقط فایل JSON خروجی را پیوست کنید؛ "
            "شامل واحدهای متصل، مشخصات واقعی Regression و شواهد قبلی لازم است."
        )
    }

    raw = (
        json.dumps(
            result, ensure_ascii=False, sort_keys=True,
            indent=2, allow_nan=False
        ) + "\n"
    ).encode("utf-8")
    digest = sha(raw)
    target = v3 / (
        "C026.R11_unit_and_regression_readiness." + digest + ".json"
    )
    if target.exists():
        need(target.read_bytes() == raw, "OUTPUT_CONTENT_CONFLICT")
    else:
        with target.open("xb") as handle:
            handle.write(raw)

    saved = target.read_bytes()
    need(sha(saved) == digest, "OUTPUT_HASH_CONFLICT")
    need(parse(saved) == result, "OUTPUT_RELOAD_CONFLICT")

    print(json.dumps({
        "status": result["status"],
        "roadmap_action": "R11",
        "historical_population": 1149,
        "reviewed_and_bound_population": len(per_case),
        "remaining_population": 1149 - len(per_case),
        "unit_accounting": totals,
        "regression_specification_count": len(regression_specs),
        "regression_tests_executed": 0,
        "relevant_existing_evidence_document_count": len(evidence_documents),
        "new_api_calls": 0,
        "metrics_calculated": False,
        "all_R11_items_complete": False,
        "write_status": "WRITTEN",
        "output": {
            "path": str(target.relative_to(root)),
            "file_bytes_sha256": digest,
            "size_bytes": len(raw),
            "reload_validation": "PASS"
        },
        "user_action_fa": result["user_action_fa"]
    }, ensure_ascii=False, indent=2))

    return result


V3_C026 = run_v3_c026()
````

````python
# C028 — R11: تعریف اصلی سناریوها و شواهد موجودِ همان پرونده‌ها
# بدون import، اجرای Runner، API، تولید پاسخ یا داوری تازه

def run_v3_c028():
    missing = [
        name for name in ("Path", "os", "json", "hashlib")
        if name not in globals()
    ]
    if missing:
        raise RuntimeError(
            "ابتدا سل imports را اجرا کنید: " + ", ".join(missing)
        )

    def need(ok, message):
        if not ok:
            raise ValueError(message)

    def sha(raw):
        return hashlib.sha256(raw).hexdigest()

    def unique_object(items):
        result = {}
        for key, value in items:
            need(key not in result, "DUPLICATE_JSON_KEY: " + key)
            result[key] = value
        return result

    def reject_constant(value):
        raise ValueError("INVALID_JSON_CONSTANT: " + value)

    def parse(raw):
        return json.loads(
            raw.decode("utf-8"),
            object_pairs_hook=unique_object,
            parse_constant=reject_constant
        )

    root = Path(os.environ.get(
        "RAG_FINANCE_ROOT",
        "/Users/macbookpro/AMIR_DATA/00_WORKS/01_PROJECTS/"
        "RAG-projects/RAG_finance"
    )).expanduser().resolve()
    out_dir = root / "reports/phase2/quality_improvement/NB22E.lab.v3"
    receipts = {}

    def read_bound(relative, digest):
        path = (root / relative).resolve()
        need(path.is_relative_to(root), "PATH_OUTSIDE_PROJECT")
        need(path.is_file(), "FILE_NOT_FOUND: " + relative)
        raw = path.read_bytes()
        need(sha(raw) == digest, "HASH_CONFLICT: " + relative)
        receipts[relative] = digest
        return parse(raw)

    def read_v3(prefix, digest):
        return read_bound(
            str((out_dir / (prefix + digest + ".json")).relative_to(root)),
            digest
        )

    sources = read_v3(
        "C027.R11_metric_and_regression_sources.",
        "900221c8c3d5f87726fbc9ab14cfb3ca74ce330f2ae69f814b509498765ee6ba"
    )
    units = read_v3(
        "C026.R11_unit_and_regression_readiness.",
        "d248ae7500dadde202ca87880f11a84cee58f57f4df7b927cc190b317dc54ee1"
    )

    inventories = [
        document for document in sources["explicit_regression_source_documents"]
        if Path(document["path"]).name.startswith("C081.source_inventory.")
        and document["status"] == "PINNED_REFERENCED_CONTENT_READ"
    ]
    need(len(inventories) == 1, "SOURCE_INVENTORY_NOT_UNIQUE")

    inventory_document = inventories[0]
    inventory = read_bound(
        inventory_document["path"],
        inventory_document["file_bytes_sha256"]
    )
    need(
        inventory == inventory_document["content"],
        "SOURCE_INVENTORY_CONTENT_CONFLICT"
    )

    required = {
        "original_regression_definitions": (
            "C078:preserved_regression_specs",
            "09b05f51fbadd8a090b14c1c2ee65036987aa8e36555354535b983ebffe02006"
        ),
        "final_issue_catalogue": (
            "C078:final_catalogue322",
            "1d1819529cabbc91a54c40b64b00a6e7219541006e99ab733fdb439289e879e3"
        ),
        "open_reviews": (
            "C078:open_reviews",
            "a7df6aff05eb1463387048930332166958deaa53cc4a46cf2b281720dfa5313a"
        )
    }

    original_documents = {}
    for name, (role, expected_sha) in required.items():
        matches = [item for item in inventory if item.get("role") == role]
        need(len(matches) == 1, "SOURCE_ROLE_NOT_UNIQUE: " + role)
        binding = matches[0]
        need(
            binding["file_bytes_sha256"] == expected_sha,
            "ORIGINAL_SOURCE_RECEIPT_CONFLICT: " + role
        )
        content = read_bound(binding["path"], expected_sha)
        original_documents[name] = {
            "path": binding["path"],
            "file_bytes_sha256": expected_sha,
            "role": role,
            "content": content
        }

    status_rows = units["regression_specifications"]
    need(
        len(status_rows) == 322
        and len({row["regression_spec_id"] for row in status_rows}) == 322,
        "REGRESSION_STATUS_POPULATION_CONFLICT"
    )
    need(
        len({row["candidate_id"] for row in status_rows}) == 258,
        "REGRESSION_CANDIDATE_COUNT_CONFLICT"
    )

    review_index = {
        row["candidate_id"]: row for row in units["per_case_unit_reviews"]
    }
    need(
        len(review_index) == len(units["per_case_unit_reviews"]) == 1047,
        "UNIT_REVIEW_POPULATION_CONFLICT"
    )

    scenario_links = []
    linked_candidate_ids = set()
    for row in status_rows:
        candidate_id = row["candidate_id"]
        has_review = candidate_id in review_index
        if has_review:
            linked_candidate_ids.add(candidate_id)
        scenario_links.append({
            "regression_spec_id": row["regression_spec_id"],
            "issue_id": row["issue_id"],
            "candidate_id": candidate_id,
            "issue_has_open_review": row["issue_has_open_review"],
            "existing_historical_review_available": has_review,
            "binding_level": "CANDIDATE_ID_ONLY",
            "expected_behavior_checked": False,
            "same_output_version_for_regression_established": False,
            "test_executed": False,
            "test_pass_granted": False
        })

    linked_scenario_count = sum(
        row["existing_historical_review_available"]
        for row in scenario_links
    )
    need(
        linked_scenario_count == 286
        and len(linked_candidate_ids) == 228,
        "INSPECTED_SCENARIO_LINK_COUNT_CONFLICT"
    )

    result = {
        "schema_version": "NB22E_V3_C028_R11_V1",
        "status": "R11_ORIGINAL_REGRESSION_DEFINITIONS_AND_MATCHED_REVIEWS_BOUND",
        "roadmap_action": "R11",
        "input_receipts": receipts,
        "original_regression_sources": original_documents,
        "historical_scenario_status_records": status_rows,
        "scenario_identity_links": scenario_links,
        "scenario_count": 322,
        "distinct_candidate_count": 258,
        "scenarios_with_existing_historical_review": linked_scenario_count,
        "distinct_candidates_with_existing_historical_review":
            len(linked_candidate_ids),
        "matched_existing_reviews": [
            review_index[candidate_id]
            for candidate_id in sorted(linked_candidate_ids)
        ],
        "scenarios_without_matched_historical_review": [
            row["regression_spec_id"] for row in scenario_links
            if not row["existing_historical_review_available"]
        ],
        "recorded_open_review_scenario_count": sum(
            row["issue_has_open_review"] for row in status_rows
        ),
        "selected_variant": units["frozen_selected_method"],
        "current_metric_readiness_preserved": sources["metric_readiness"],
        "candidate_id_match_is_regression_pass": False,
        "original_specification_schema_not_guessed": True,
        "actual_regression_tests_executed": 0,
        "new_api_calls": 0,
        "new_model_pairs": 0,
        "new_semantic_judgments": 0,
        "quality_metrics_recomputed": False,
        "historical_files_modified": False,
        "gate_pass_granted": False,
        "all_R11_items_complete": False,
        "advance_to_next_roadmap_action": False,
        "next_step": (
            "READ_ORIGINAL_EXPECTATIONS_AND_DETERMINE_EXACT_EXECUTABLE_"
            "CHECKS_WITH_EXISTING_VERSION_MATCHED_OUTPUTS"
        ),
        "user_action_fa": (
            "فقط فایل JSON خروجی را پیوست کنید؛ این بار تعریف اصلی "
            "سناریوها، کاتالوگ و آرای موجودِ پرونده‌های مرتبط داخل بسته است."
        )
    }

    raw = (
        json.dumps(
            result, ensure_ascii=False, sort_keys=True,
            indent=2, allow_nan=False
        ) + "\n"
    ).encode("utf-8")
    digest = sha(raw)
    target = out_dir / (
        "C028.R11_original_regression_definitions." + digest + ".json"
    )

    if target.exists():
        need(target.read_bytes() == raw, "OUTPUT_CONTENT_CONFLICT")
    else:
        with target.open("xb") as handle:
            handle.write(raw)

    saved = target.read_bytes()
    need(sha(saved) == digest, "OUTPUT_HASH_CONFLICT")
    need(parse(saved) == result, "OUTPUT_RELOAD_CONFLICT")

    print(json.dumps({
        "status": result["status"],
        "roadmap_action": "R11",
        "scenario_count": 322,
        "distinct_candidate_count": 258,
        "scenarios_with_existing_historical_review": linked_scenario_count,
        "distinct_candidates_with_existing_historical_review":
            len(linked_candidate_ids),
        "original_source_files": [
            {
                "role": item["role"],
                "path": item["path"],
                "file_bytes_sha256": item["file_bytes_sha256"]
            }
            for item in original_documents.values()
        ],
        "actual_regression_tests_executed": 0,
        "new_api_calls": 0,
        "all_R11_items_complete": False,
        "write_status": "WRITTEN",
        "output": {
            "path": str(target.relative_to(root)),
            "file_bytes_sha256": digest,
            "size_bytes": len(raw),
            "reload_validation": "PASS"
        },
        "user_action_fa": result["user_action_fa"]
    }, ensure_ascii=False, indent=2))
    return result


V3_C028 = run_v3_c028()
````

### کار یازده — ثبت بررسی‌های آفلاین Regression

هویت ۳۲۲ تعریف سناریو با کاتالوگ اصلی تطبیق یافت.
۲۶ بررسی حفظ وضعیت تاریخی با موفقیت اجرا شد.

۲۹۶ سناریو برای بررسی رفتار نسخهٔ منتخب به شاهد هم‌نسخه نیاز دارند:
۲۶۰ بازیابی، ۳۱ رفتار پاسخ و ۵ شاهد استناد.

نتایج آفلاین در دامنهٔ دادهٔ تاریخی ثبت می‌شوند.
وضعیت Gate محصول تا احراز پیش‌نیازهای نسخهٔ منتخب باز می‌ماند.

</div>

````python
# C029 — R11: اجرای بررسی‌های حفظ وضعیت و ثبت موانع هر سناریو
# وابسته به imports سل اول

def run_v3_c029():
    root = Path(os.environ.get(
        "RAG_FINANCE_ROOT",
        "/Users/macbookpro/AMIR_DATA/00_WORKS/01_PROJECTS/"
        "RAG-projects/RAG_finance"
    )).expanduser().resolve()
    directory = root / "reports/phase2/quality_improvement/NB22E.lab.v3"

    source_sha = "9cd26651bf9794b4ef993d899bdf844c113930e96c43a0a6feffbca540565c08"
    source_path = directory / (
        "C028.R11_original_regression_definitions." + source_sha + ".json"
    )

    def need(condition, message):
        if not condition:
            raise ValueError(message)

    raw = source_path.read_bytes()
    need(hashlib.sha256(raw).hexdigest() == source_sha, "C028_HASH_CONFLICT")
    source = json.loads(raw)
    documents = source["original_regression_sources"]
    specs = documents["original_regression_definitions"]["content"]
    catalogue = documents["final_issue_catalogue"]["content"]
    issues = {row["issue_id"]: row for row in catalogue}

    need(len(specs) == len(catalogue) == len(issues) == 322,
         "REGRESSION_POPULATION_CONFLICT")
    need(len({row["regression_spec_id"] for row in specs}) == 322,
         "DUPLICATE_REGRESSION_SPEC")

    blockers = {
        "FIXED_QUERY_RETRIEVAL_REGRESSION":
            "SAME_POOL50_SELECTED_VERSION_RANKING_AND_REFERENCE_MAPPING_REQUIRED",
        "SOURCE_BOUND_ANSWER_DECISION_REGRESSION":
            "VERSION_MATCHED_RESPONSE_AND_SCOPED_BEHAVIOR_ASSESSMENT_REQUIRED",
        "CITATION_WITNESS_RECORDING_REGRESSION":
            "VERSION_MATCHED_LINK_WITNESS_OR_EXPLICIT_LINK_VERDICT_REQUIRED"
    }

    outcomes = []
    for spec in specs:
        issue = issues[spec["issue_id"]]
        spec_id = spec["regression_spec_id"]
        need(
            issue["candidate_id"] == spec["candidate_id"]
            and issue["regression_spec_id"] == spec_id
            and issue["response_document_sha256"]
            == spec["frozen_response_document_sha256"],
            spec_id + ": IDENTITY_OR_RESPONSE_VERSION_CONFLICT"
        )

        kind = spec["kind"]
        witness = issue["witness"]
        row = {
            "regression_spec_id": spec_id,
            "issue_id": spec["issue_id"],
            "candidate_id": spec["candidate_id"],
            "kind": kind,
            "response_document_sha256": issue["response_document_sha256"],
            "original_assertion": spec["assertion"]
        }

        if kind in blockers:
            row.update({
                "status": "BLOCKED",
                "reason": blockers[kind],
                "required_inputs_as_recorded": spec.get("required_inputs"),
                "product_test_executed": False
            })
        else:
            if kind == "EVALUATION_ACCOUNTING_REGRESSION":
                passed = (
                    issue["status"] == "UNRESOLVED_NOT_MODEL_ERROR"
                    and "recorded_action26_output" in witness
                    and witness["recorded_action26_output"] is None
                    and spec["automatic_resend_permitted"] is False
                )
            elif kind == "UNDETERMINED_DECISION_PRESERVATION":
                passed = (
                    issue["status"] == "UNDETERMINED_NOT_AUTOMATICALLY_INCORRECT"
                    and witness["decision_assessment"] == "UNDETERMINED"
                )
            elif kind == "DISPUTE_PRESERVATION_AND_REVIEW":
                passed = (
                    issue["status"] == "DISPUTED_NOT_CONFIRMED_VIOLATION"
                    and witness["recorded_model_label_preserved"] == "RISK_IDENTIFIED"
                    and witness["reporting_status"] == "DISPUTED_MODEL_RISK_JUDGMENT"
                )
            else:
                raise ValueError("UNKNOWN_REGRESSION_KIND: " + kind)

            row.update({
                "status": "PASS" if passed else "FAIL",
                "scope": "OFFLINE_HISTORICAL_STATE_PRESERVATION",
                "product_test_executed": False
            })

        outcomes.append(row)

    counts = {
        status: sum(row["status"] == status for row in outcomes)
        for status in ("PASS", "FAIL", "BLOCKED")
    }
    need(counts == {"PASS": 26, "FAIL": 0, "BLOCKED": 296},
         "OFFLINE_REPLAY_RESULT_CONFLICT: " + str(counts))

    result = {
        "schema_version": "NB22E_V3_C029_R11_V1",
        "status": "R11_OFFLINE_PRESERVATION_CHECKS_COMPLETE_PRODUCT_REGRESSION_OPEN",
        "roadmap_action": "R11",
        "input_receipt": {
            "path": str(source_path.relative_to(root)),
            "file_bytes_sha256": source_sha
        },
        "selected_variant": source["selected_variant"],
        "scenario_count": len(outcomes),
        "offline_check_status_counts": counts,
        "offline_checks_executed": counts["PASS"] + counts["FAIL"],
        "product_regression_tests_executed": 0,
        "scenario_outcomes": outcomes,
        "new_api_calls": 0,
        "new_semantic_judgments": 0,
        "gate_pass_granted": False,
        "all_R11_items_complete": False,
        "next_step": "BIND_SELECTED_VERSION_OUTPUTS_TO_BLOCKED_SCENARIO_ASSERTIONS"
    }

    payload = (
        json.dumps(result, ensure_ascii=False, sort_keys=True,
                   indent=2, allow_nan=False) + "\n"
    ).encode("utf-8")
    digest = hashlib.sha256(payload).hexdigest()
    output_path = directory / ("C029.R11_regression_disposition." + digest + ".json")
    try:
        with output_path.open("xb") as handle:
            handle.write(payload)
    except FileExistsError:
        need(output_path.read_bytes() == payload, "OUTPUT_CONTENT_CONFLICT")

    saved = output_path.read_bytes()
    need(hashlib.sha256(saved).hexdigest() == digest
         and json.loads(saved) == result, "OUTPUT_RELOAD_CONFLICT")

    print(json.dumps({
        "status": result["status"],
        "offline_check_status_counts": counts,
        "offline_checks_executed": result["offline_checks_executed"],
        "product_regression_tests_executed": 0,
        "all_R11_items_complete": False,
        "output": {
            "path": str(output_path.relative_to(root)),
            "file_bytes_sha256": digest,
            "reload_validation": "PASS"
        }
    }, ensure_ascii=False, indent=2))
    return result


V3_C029 = run_v3_c029()
````

### کار یازده — تطبیق سناریوهای بازیابی با رتبه‌های نسخهٔ منتخب

۲۶۰ سناریوی بازیابی به ۲۲۶ سؤال Test تاریخی مربوط‌اند.
فایل رتبه‌های منتخب char_rrf با هش ثبت‌شده در فریز کار ده خوانده می‌شود.

برای هر سناریو، وجود شناسهٔ سؤال و split رکورد رتبه بررسی می‌شود.
سناریویی که ورودی هم‌دامنه ندارد، با دلیل دقیق BLOCKED می‌ماند.
در صورت تطبیق شناسه و split، بررسی Pool50 و نگاشت مرجع همچنان لازم است.

نتیجهٔ ۲۶ بررسی آفلاین قبلی محفوظ می‌ماند.
اجرای این سل، بررسی پیش‌نیاز Regression است.
کار یازده همچنان باز می‌ماند.

````python
# C030 — R11: تطبیق ۲۶۰ سناریوی بازیابی با فایل فریز‌شدهٔ char_rrf
# وابسته به imports سل اول

def run_v3_c030():
    missing = [
        name for name in ("Path", "os", "json", "hashlib")
        if name not in globals()
    ]
    if missing:
        raise RuntimeError("ابتدا سل imports را اجرا کنید: " + ", ".join(missing))

    def need(ok, message):
        if not ok:
            raise ValueError(message)

    def sha(raw):
        return hashlib.sha256(raw).hexdigest()

    root = Path(os.environ.get(
        "RAG_FINANCE_ROOT",
        "/Users/macbookpro/AMIR_DATA/00_WORKS/01_PROJECTS/"
        "RAG-projects/RAG_finance"
    )).expanduser().resolve()
    directory = root / "reports/phase2/quality_improvement/NB22E.lab.v3"
    receipts = {}

    def read_pinned(prefix, digest):
        path = directory / (prefix + digest + ".json")
        raw = path.read_bytes()
        need(sha(raw) == digest, "HASH_CONFLICT: " + path.name)
        receipts[prefix] = {
            "path": str(path.relative_to(root)),
            "file_bytes_sha256": digest
        }
        return json.loads(raw)

    frozen = read_pinned(
        "C024.R10_selected_evidence_freeze.",
        "b807237f302b824c5ad943b5e3adb68a79ebd5c29c51c30edc488b88deec82e8"
    )
    definitions = read_pinned(
        "C028.R11_original_regression_definitions.",
        "9cd26651bf9794b4ef993d899bdf844c113930e96c43a0a6feffbca540565c08"
    )
    previous = read_pinned(
        "C029.R11_regression_disposition.",
        "0053d51b679d38384e80d9932e6e608e87b66ff24fd45f64e0f63256d9a3759f"
    )
    need(
        previous["input_receipt"]["file_bytes_sha256"]
        == receipts["C028.R11_original_regression_definitions."]["file_bytes_sha256"],
        "C029_C028_BINDING_CONFLICT"
    )

    selected = frozen["selected_method"]
    need(
        selected["variant"] == previous["selected_variant"] == "char_rrf",
        "SELECTED_VARIANT_CONFLICT"
    )

    documents = definitions["original_regression_sources"]
    catalogue = documents["final_issue_catalogue"]["content"]
    issue_index = {row["issue_id"]: row for row in catalogue}
    specs = [
        row for row in documents["original_regression_definitions"]["content"]
        if row["kind"] == "FIXED_QUERY_RETRIEVAL_REGRESSION"
    ]
    wanted = {row["candidate_id"] for row in specs}
    need(len(specs) == 260 and len(wanted) == 226, "SCENARIO_COUNT_CONFLICT")

    binding = selected["ranking_artifact"]
    rank_path = (root / binding["path"]).resolve()
    need(rank_path.is_relative_to(root), "RANK_PATH_OUTSIDE_PROJECT")

    matching_rows = {}
    split_counts = {}
    ranking_ids = set()
    digest = hashlib.sha256()
    before = rank_path.stat()

    with rank_path.open("rb") as stream:
        for line in stream:
            digest.update(line)
            row = json.loads(line)
            candidate_id = row["candidate_id"]
            split = row["split"]
            need(candidate_id not in ranking_ids, "DUPLICATE_RANK_CANDIDATE")
            need(row["variant"] == selected["variant"], "RANK_VARIANT_CONFLICT")
            ranking_ids.add(candidate_id)
            split_counts[split] = split_counts.get(split, 0) + 1
            if candidate_id in wanted:
                matching_rows[candidate_id] = split

    after = rank_path.stat()
    need(
        (before.st_size, before.st_mtime_ns, before.st_ino)
        == (after.st_size, after.st_mtime_ns, after.st_ino),
        "RANK_FILE_CHANGED_DURING_READ"
    )
    need(
        digest.hexdigest() == binding["file_bytes_sha256"],
        "RANK_FILE_HASH_CONFLICT"
    )

    outcomes = []
    for spec in specs:
        candidate_id = spec["candidate_id"]
        issue = issue_index[spec["issue_id"]]
        need(
            issue["candidate_id"] == candidate_id
            and issue["regression_spec_id"] == spec["regression_spec_id"],
            "SCENARIO_IDENTITY_CONFLICT"
        )
        population = issue["witness"]["original_observation"]["population"]
        need(population == "HISTORICAL_TEST", "SCENARIO_POPULATION_CONFLICT")

        matched_split = matching_rows.get(candidate_id)
        if matched_split is None:
            reason = "CANDIDATE_ABSENT_FROM_FROZEN_SELECTED_RANKINGS"
        elif matched_split != "test":
            reason = "SELECTED_RANK_ROW_SPLIT_DIFFERS_FROM_HISTORICAL_TEST"
        else:
            reason = "EXACT_POOL50_AND_REFERENCE_BINDING_REQUIRED"

        outcomes.append({
            "regression_spec_id": spec["regression_spec_id"],
            "candidate_id": candidate_id,
            "required_population": population,
            "selected_rank_row_found": matched_split is not None,
            "selected_rank_row_split": matched_split,
            "status": "BLOCKED",
            "reason": reason,
            "required_inputs": spec["required_inputs"],
            "product_test_executed": False
        })

    reason_counts = {}
    for row in outcomes:
        reason = row["reason"]
        reason_counts[reason] = reason_counts.get(reason, 0) + 1

    result = {
        "schema_version": "NB22E_V3_C030_R11_V1",
        "status": "R11_RETRIEVAL_REGRESSION_INPUT_COMPATIBILITY_CHECKED",
        "roadmap_action": "R11",
        "input_receipts": receipts,
        "selected_variant": selected["variant"],
        "selected_ranking_receipt": binding,
        "selected_ranking_split_counts": split_counts,
        "retrieval_scenario_count": len(outcomes),
        "distinct_retrieval_question_count": len(wanted),
        "matching_candidate_count": len(matching_rows),
        "retrieval_blocker_counts": reason_counts,
        "retrieval_scenario_outcomes": outcomes,
        "prior_offline_check_status_counts":
            previous["offline_check_status_counts"],
        "product_regression_tests_executed": 0,
        "new_api_calls": 0,
        "gate_pass_granted": False,
        "all_R11_items_complete": False,
        "next_step": "BIND_SELECTED_TEST_POOL50_RANKS_OR_RECORD_THE_PRECISE_GAP"
    }

    payload = (
        json.dumps(result, ensure_ascii=False, sort_keys=True,
                   indent=2, allow_nan=False) + "\n"
    ).encode("utf-8")
    output_sha = sha(payload)
    output_path = directory / (
        "C030.R11_retrieval_input_compatibility." + output_sha + ".json"
    )

    try:
        with output_path.open("xb") as handle:
            handle.write(payload)
    except FileExistsError:
        need(output_path.read_bytes() == payload, "OUTPUT_CONTENT_CONFLICT")

    saved = output_path.read_bytes()
    need(
        sha(saved) == output_sha and json.loads(saved) == result,
        "OUTPUT_RELOAD_CONFLICT"
    )

    print(json.dumps({
        "status": result["status"],
        "selected_ranking_split_counts": split_counts,
        "retrieval_scenario_count": len(outcomes),
        "distinct_retrieval_question_count": len(wanted),
        "matching_candidate_count": len(matching_rows),
        "retrieval_blocker_counts": reason_counts,
        "product_regression_tests_executed": 0,
        "all_R11_items_complete": False,
        "output": {
            "path": str(output_path.relative_to(root)),
            "file_bytes_sha256": output_sha,
            "reload_validation": "PASS"
        }
    }, ensure_ascii=False, indent=2))
    return result


V3_C030 = run_v3_c030()
````

````python
# C031 — R11: اتصال شاهدهای تاریخی و ثبت پیش‌نیاز تست نسخهٔ منتخب

def run_v3_c031():
    root = Path(os.environ.get(
        "RAG_FINANCE_ROOT",
        "/Users/macbookpro/AMIR_DATA/00_WORKS/01_PROJECTS/"
        "RAG-projects/RAG_finance"
    )).expanduser().resolve()
    directory = root / "reports/phase2/quality_improvement/NB22E.lab.v3"

    def need(condition, message):
        if not condition:
            raise ValueError(message)

    receipts = {}

    def read_pinned(stem, digest):
        path = directory / (stem + "." + digest + ".json")
        raw = path.read_bytes()
        need(hashlib.sha256(raw).hexdigest() == digest, "HASH_CONFLICT: " + stem)
        receipts[stem] = {
            "path": str(path.relative_to(root)),
            "file_bytes_sha256": digest
        }
        return json.loads(raw)

    source = read_pinned(
        "C028.R11_original_regression_definitions",
        "9cd26651bf9794b4ef993d899bdf844c113930e96c43a0a6feffbca540565c08"
    )
    prior = read_pinned(
        "C029.R11_regression_disposition",
        "0053d51b679d38384e80d9932e6e608e87b66ff24fd45f64e0f63256d9a3759f"
    )
    retrieval = read_pinned(
        "C030.R11_retrieval_input_compatibility",
        "db6156545000816b4ee2eda5a7c2a09b898348be5dcb38330508c2db69c93ab8"
    )
    need(prior["offline_check_status_counts"] ==
         {"PASS": 26, "FAIL": 0, "BLOCKED": 296}, "C029_COUNTS_CONFLICT")
    need(retrieval["retrieval_blocker_counts"] ==
         {"CANDIDATE_ABSENT_FROM_FROZEN_SELECTED_RANKINGS": 260},
         "C030_COUNTS_CONFLICT")

    documents = source["original_regression_sources"]
    specs = documents["original_regression_definitions"]["content"]
    issues = {
        row["issue_id"]: row
        for row in documents["final_issue_catalogue"]["content"]
    }
    reviews = {
        row["candidate_id"]: row for row in source["matched_existing_reviews"]
    }

    answer_rows = []
    citation_rows = []

    for spec in specs:
        kind = spec["kind"]
        if kind not in {
            "SOURCE_BOUND_ANSWER_DECISION_REGRESSION",
            "CITATION_WITNESS_RECORDING_REGRESSION"
        }:
            continue

        candidate_id = spec["candidate_id"]
        spec_id = spec["regression_spec_id"]
        response_sha = spec["frozen_response_document_sha256"]
        issue = issues[spec["issue_id"]]
        need(
            issue["candidate_id"] == candidate_id
            and issue["regression_spec_id"] == spec_id
            and issue["response_document_sha256"] == response_sha,
            spec_id + ": HISTORICAL_IDENTITY_CONFLICT"
        )
        existing = reviews.get(candidate_id)
        same_response = (
            existing is not None
            and existing["response_document_sha256"] == response_sha
        )
        row = {
            "regression_spec_id": spec_id,
            "candidate_id": candidate_id,
            "historical_response_sha256": response_sha,
            "historical_A07_response_hash_matches": same_response,
            "selected_version_test_status": "BLOCKED"
        }

        if kind == "SOURCE_BOUND_ANSWER_DECISION_REGRESSION":
            row.update({
                "expected_behavior": spec["expected_behavior"],
                "historical_observed_behavior": spec["historical_observed_behavior"],
                "historical_decision_assessment":
                    issue["witness"]["decision_assessment"],
                "required_limit_disclosed_as_recorded":
                    spec["required_limit_disclosed"],
                "blocker": "SELECTED_VERSION_RESPONSE_AND_BEHAVIOR_REVIEW_REQUIRED"
            })
            answer_rows.append(row)
            continue

        need(same_response, spec_id + ": A07_RESPONSE_HASH_CONFLICT")
        witness = issue["witness"]
        claim_index = witness["claim_index"]
        evidence_id = witness["evidence_id"]
        matched_links = [
            link for link in existing["review"]["citation_links"]
            if link["claim_index"] == claim_index
            and link["evidence_id"] == evidence_id
        ]
        need(len(matched_links) == 1, spec_id + ": LINK_NOT_UNIQUE")
        link = matched_links[0]
        support = link["support"]
        need(support in {"SUPPORTED", "NOT_SUPPORTED"}, spec_id + ": SUPPORT_CHANGED")

        spans = []
        if support == "SUPPORTED":
            evidence_text = witness["evidence_text"]
            need(bool(link["evidence_quotes"]), spec_id + ": QUOTES_MISSING")
            for quote in link["evidence_quotes"]:
                need(isinstance(quote, str) and bool(quote),
                     spec_id + ": INVALID_QUOTE")
                start = evidence_text.find(quote)
                need(start >= 0, spec_id + ": QUOTE_NOT_FOUND")
                while start >= 0:
                    spans.append({
                        "start": start,
                        "end": start + len(quote),
                        "quote": quote
                    })
                    start = evidence_text.find(quote, start + 1)

        row.update({
            "claim_index": claim_index,
            "evidence_id": evidence_id,
            "historical_model_link_support": support,
            "historical_exact_spans": spans,
            "span_unit": "UNICODE_CODE_POINT",
            "span_scope": "C078_RECORDED_EVIDENCE_TEXT",
            "historical_review_source": existing["provenance"],
            "independent_source_locator_verified": False,
            "blocker": "SELECTED_VERSION_RESPONSE_AND_QUALIFIED_LINK_EVIDENCE_REQUIRED"
        })
        citation_rows.append(row)

    answer_matches = sum(
        row["historical_A07_response_hash_matches"] for row in answer_rows
    )
    supported = sum(
        row["historical_model_link_support"] == "SUPPORTED" for row in citation_rows
    )
    need(
        (len(answer_rows), answer_matches, len(citation_rows), supported)
        == (31, 26, 5, 4),
        "INSPECTED_COUNTS_CONFLICT"
    )

    result = {
        "schema_version": "NB22E_V3_C031_R11_V1",
        "status": "R11_ANSWER_AND_CITATION_HISTORICAL_EVIDENCE_BOUND",
        "roadmap_action": "R11",
        "input_receipts": receipts,
        "selected_variant": retrieval["selected_variant"],
        "answer_scenarios": answer_rows,
        "citation_scenarios": citation_rows,
        "counts": {
            "answer_scenarios": len(answer_rows),
            "answer_A07_hash_matches": answer_matches,
            "citation_scenarios": len(citation_rows),
            "historical_supported_with_exact_quotes": supported,
            "historical_not_supported": len(citation_rows) - supported,
            "prior_offline_preservation_passes": 26,
            "selected_version_regressions_blocked": 260 + len(answer_rows) + len(citation_rows)
        },
        "product_regression_tests_executed": 0,
        "new_api_calls": 0,
        "new_semantic_judgments": 0,
        "gate_pass_granted": False,
        "all_R11_items_complete": False
    }

    payload = (
        json.dumps(result, ensure_ascii=False, sort_keys=True,
                   indent=2, allow_nan=False) + "\n"
    ).encode("utf-8")
    digest = hashlib.sha256(payload).hexdigest()
    output = directory / ("C031.R11_answer_citation_binding." + digest + ".json")
    try:
        with output.open("xb") as handle:
            handle.write(payload)
    except FileExistsError:
        need(output.read_bytes() == payload, "OUTPUT_CONTENT_CONFLICT")

    saved = output.read_bytes()
    need(
        hashlib.sha256(saved).hexdigest() == digest
        and json.loads(saved) == result,
        "OUTPUT_RELOAD_CONFLICT"
    )
    print(json.dumps({
        "status": result["status"],
        "counts": result["counts"],
        "product_regression_tests_executed": 0,
        "all_R11_items_complete": False,
        "output": {
            "path": str(output.relative_to(root)),
            "file_bytes_sha256": digest,
            "reload_validation": "PASS"
        }
    }, ensure_ascii=False, indent=2))
    return result


V3_C031 = run_v3_c031()
````

````python
# C032 — R11: بررسی سازگاری خروجی‌های متصل با نسخهٔ منتخب
# با imports موجود در سل اول

def run_v3_c032():
    required = ("Path", "os", "json", "hashlib")
    missing = [name for name in required if name not in globals()]
    if missing:
        raise RuntimeError("ابتدا سل imports را اجرا کنید: " + ", ".join(missing))

    root = Path(os.environ.get(
        "RAG_FINANCE_ROOT",
        "/Users/macbookpro/AMIR_DATA/00_WORKS/01_PROJECTS/"
        "RAG-projects/RAG_finance"
    )).expanduser().resolve()
    directory = root / "reports/phase2/quality_improvement/NB22E.lab.v3"
    receipts = {}

    def need(condition, message):
        if not condition:
            raise ValueError(message)

    def sha(raw):
        return hashlib.sha256(raw).hexdigest()

    def read_pinned(stem, digest):
        path = directory / (stem + "." + digest + ".json")
        raw = path.read_bytes()
        need(sha(raw) == digest, "HASH_CONFLICT: " + stem)
        receipts[stem.split(".")[0]] = {
            "path": str(path.relative_to(root)),
            "file_bytes_sha256": digest
        }
        return json.loads(raw)

    frozen = read_pinned(
        "C024.R10_selected_evidence_freeze",
        "b807237f302b824c5ad943b5e3adb68a79ebd5c29c51c30edc488b88deec82e8"
    )
    units = read_pinned(
        "C026.R11_unit_and_regression_readiness",
        "d248ae7500dadde202ca87880f11a84cee58f57f4df7b927cc190b317dc54ee1"
    )
    preservation = read_pinned(
        "C029.R11_regression_disposition",
        "0053d51b679d38384e80d9932e6e608e87b66ff24fd45f64e0f63256d9a3759f"
    )
    retrieval = read_pinned(
        "C030.R11_retrieval_input_compatibility",
        "db6156545000816b4ee2eda5a7c2a09b898348be5dcb38330508c2db69c93ab8"
    )
    responses = read_pinned(
        "C031.R11_answer_citation_binding",
        "4950b86eb86e688f840f2d2a464b054e2613c13a2e2872bed137825e3da67b6a"
    )

    selected = frozen["selected_method"]
    variant = selected["variant"]
    need(
        variant == units["frozen_selected_method"]
        == preservation["selected_variant"]
        == retrieval["selected_variant"]
        == responses["selected_variant"] == "char_rrf",
        "SELECTED_VARIANT_CONFLICT"
    )
    need(
        retrieval["selected_ranking_receipt"] == selected["ranking_artifact"],
        "SELECTED_RANKING_RECEIPT_CONFLICT"
    )
    need(
        responses["input_receipts"]["C030.R11_retrieval_input_compatibility"]
        == receipts["C030"],
        "C031_C030_BINDING_CONFLICT"
    )

    historical_tables = [
        doc for doc in units["relevant_existing_evidence_documents"]
        if Path(doc["path"]).name.startswith("C069.metric_table.")
    ]
    need(len(historical_tables) == 1, "HISTORICAL_METRIC_TABLE_NOT_UNIQUE")
    table = historical_tables[0]
    historical_rows = table["stored_content"]
    historical_recall = [
        row for row in historical_rows if row["metric"] == "recall_at_k"
    ]
    need(len(historical_recall) == 1, "HISTORICAL_RECALL_ROW_NOT_UNIQUE")
    historical_rank_source = historical_recall[0]
    need(
        historical_rank_source["population"] == "HISTORICAL_TEST",
        "HISTORICAL_RETRIEVAL_POPULATION_CONFLICT"
    )

    need(
        retrieval["selected_ranking_split_counts"] == {"development": 2664}
        and retrieval["matching_candidate_count"] == 0,
        "C030_SPLIT_OR_MATCH_COUNT_CONFLICT"
    )
    need(
        units["selected_method_answer_quality_established"] is False,
        "SELECTED_RESPONSE_EVIDENCE_STATE_CHANGED"
    )

    def index_scenarios(rows):
        indexed = {row["regression_spec_id"]: row for row in rows}
        need(len(indexed) == len(rows), "DUPLICATE_SCENARIO_ID")
        return indexed

    prior = index_scenarios(preservation["scenario_outcomes"])
    rank_cases = index_scenarios(retrieval["retrieval_scenario_outcomes"])
    answer_cases = index_scenarios(responses["answer_scenarios"])
    citation_cases = index_scenarios(responses["citation_scenarios"])

    blocked_ids = {
        spec_id for spec_id, row in prior.items()
        if row["status"] == "BLOCKED"
    }
    need(
        len(prior) == 322
        and preservation["offline_check_status_counts"]
        == {"PASS": 26, "FAIL": 0, "BLOCKED": 296},
        "C029_POPULATION_CONFLICT"
    )
    need(
        (len(rank_cases), len(answer_cases), len(citation_cases)) == (260, 31, 5)
        and len(set(rank_cases) | set(answer_cases) | set(citation_cases)) == 296
        and set(rank_cases) | set(answer_cases) | set(citation_cases) == blocked_ids,
        "BLOCKED_SCENARIO_SET_CONFLICT"
    )

    outcomes = []
    for kind, rows in (
        ("RETRIEVAL", rank_cases),
        ("ANSWER_DECISION", answer_cases),
        ("CITATION_WITNESS", citation_cases)
    ):
        for spec_id, row in sorted(rows.items()):
            need(
                row["candidate_id"] == prior[spec_id]["candidate_id"],
                spec_id + ": CANDIDATE_BINDING_CONFLICT"
            )

            if kind == "RETRIEVAL":
                need(
                    row["selected_rank_row_found"] is False
                    and row["required_population"] == "HISTORICAL_TEST",
                    spec_id + ": RETRIEVAL_SCOPE_CONFLICT"
                )
                blocker = "SELECTED_TEST_QUERY_RANKS_ABSENT_FROM_FROZEN_ARTIFACT"
                required_input = (
                    "Selected-version ranks bound to this query version, "
                    "exact Pool50 and historical reference mapping."
                )
            else:
                need(
                    row["historical_response_sha256"]
                    == prior[spec_id]["response_document_sha256"],
                    spec_id + ": HISTORICAL_RESPONSE_HASH_CONFLICT"
                )
                blocker = "SELECTED_VERSION_RESPONSE_NOT_ESTABLISHED_IN_BOUND_EVIDENCE"
                required_input = (
                    "Selected-version response, matching evidence and "
                    "source-bound assessment of the expected behavior."
                    if kind == "ANSWER_DECISION"
                    else
                    "Selected-version response and qualified witness or "
                    "explicit verdict for the exact claim-evidence link."
                )

            outcomes.append({
                "regression_spec_id": spec_id,
                "candidate_id": row["candidate_id"],
                "kind": kind,
                "status": "BLOCKED",
                "blocker": blocker,
                "required_input": required_input,
                "original_assertion": prior[spec_id]["original_assertion"],
                "bound_evidence": row,
                "product_test_executed": False
            })

    result = {
        "schema_version": "NB22E_V3_C032_R11_V1",
        "status": "R11_BOUND_OUTPUT_COMPATIBILITY_ASSESSED",
        "roadmap_action": "R11",
        "selected_variant": variant,
        "input_receipts": receipts,
        "scope": "EXPLICIT_BOUND_SOURCES_ONLY",
        "selected_ranking_receipt": selected["ranking_artifact"],
        "historical_retrieval_source": {
            "metric_table_path": table["path"],
            "population": historical_rank_source["population"],
            "ranking_path_as_recorded": historical_rank_source["source"]
        },
        "projectwide_artifact_absence_established": False,
        "scenario_counts": {
            "historical_preservation_pass": 26,
            "selected_retrieval_blocked": len(rank_cases),
            "selected_answer_decision_blocked": len(answer_cases),
            "selected_citation_witness_blocked": len(citation_cases)
        },
        "selected_version_scenario_outcomes": outcomes,
        "product_regression_tests_executed": 0,
        "new_api_calls": 0,
        "new_semantic_judgments": 0,
        "gate_pass_granted": False,
        "all_R11_items_complete": False,
        "advance_to_next_roadmap_action": False,
        "next_step": "RESOLVE_SELECTED_TEST_RANK_AND_RESPONSE_PROVENANCE"
    }

    payload = (
        json.dumps(result, ensure_ascii=False, sort_keys=True,
                   indent=2, allow_nan=False) + "\n"
    ).encode("utf-8")
    digest = sha(payload)
    output = directory / ("C032.R11_bound_output_compatibility." + digest + ".json")

    try:
        with output.open("xb") as handle:
            handle.write(payload)
    except FileExistsError:
        need(output.read_bytes() == payload, "OUTPUT_CONTENT_CONFLICT")

    saved = output.read_bytes()
    need(sha(saved) == digest and json.loads(saved) == result,
         "OUTPUT_RELOAD_CONFLICT")

    print(json.dumps({
        "status": result["status"],
        "scope": result["scope"],
        "scenario_counts": result["scenario_counts"],
        "historical_retrieval_source": result["historical_retrieval_source"],
        "product_regression_tests_executed": 0,
        "all_R11_items_complete": False,
        "output": {
            "path": str(output.relative_to(root)),
            "file_bytes_sha256": digest,
            "reload_validation": "PASS"
        }
    }, ensure_ascii=False, indent=2))
    return result


V3_C032 = run_v3_c032()
````

````python
# C033 — R11: جمع‌بندی سه بند باقی‌مانده در یک گزارش
# وابسته به imports سل اول

def run_v3_c033():
    root = Path(os.environ.get(
        "RAG_FINANCE_ROOT",
        "/Users/macbookpro/AMIR_DATA/00_WORKS/01_PROJECTS/"
        "RAG-projects/RAG_finance"
    )).expanduser().resolve()
    directory = root / "reports/phase2/quality_improvement/NB22E.lab.v3"
    receipts = {}

    def need(condition, message):
        if not condition:
            raise ValueError(message)

    def read_pinned(stem, digest):
        path = directory / (stem + "." + digest + ".json")
        raw = path.read_bytes()
        need(hashlib.sha256(raw).hexdigest() == digest, "HASH_CONFLICT: " + stem)
        receipts[stem.split(".")[0]] = {
            "path": str(path.relative_to(root)),
            "file_bytes_sha256": digest
        }
        return json.loads(raw)

    frozen = read_pinned(
        "C024.R10_selected_evidence_freeze",
        "b807237f302b824c5ad943b5e3adb68a79ebd5c29c51c30edc488b88deec82e8"
    )
    readiness = read_pinned(
        "C027.R11_metric_and_regression_sources",
        "900221c8c3d5f87726fbc9ab14cfb3ca74ce330f2ae69f814b509498765ee6ba"
    )
    preservation = read_pinned(
        "C029.R11_regression_disposition",
        "0053d51b679d38384e80d9932e6e608e87b66ff24fd45f64e0f63256d9a3759f"
    )
    compatibility = read_pinned(
        "C032.R11_bound_output_compatibility",
        "1768832db7d097c82b3595c706e0b7f64800eb1bd1fd970da4de813b34e9d635"
    )

    selected = frozen["selected_method"]["variant"]
    need(
        selected == compatibility["selected_variant"]
        == preservation["selected_variant"] == "char_rrf",
        "SELECTED_VARIANT_CONFLICT"
    )
    need(
        compatibility["input_receipts"]["C024"] == receipts["C024"]
        and compatibility["input_receipts"]["C029"] == receipts["C029"],
        "COMPATIBILITY_INPUT_CONFLICT"
    )

    historical = readiness["historical_gate_preserved"]
    historical_metrics = historical["full_metric_matrix_preserved"]
    readiness_by_metric = {
        row["metric"]: row for row in readiness["metric_readiness"]
    }
    need(
        len(historical_metrics) == 16
        and len({row["metric"] for row in historical_metrics}) == 16
        and len(readiness_by_metric) == 12,
        "METRIC_POPULATION_CONFLICT"
    )

    selected_metrics = []
    for row in historical_metrics:
        metric = row["metric"]
        if row["historical_status"] == "BLOCKED":
            evidence = readiness_by_metric[metric]
            need(
                evidence["frozen_policy_preserved"] == row["frozen_policy"],
                "FROZEN_POLICY_CONFLICT: " + metric
            )
            reason = evidence["reason_fa"]
        else:
            need(row["historical_status"] in {"PASS", "FAIL"},
                 "UNKNOWN_HISTORICAL_STATUS")
            reason = (
                "نتیجهٔ موجود مربوط به Test تاریخی است. برای تصمیم این سنجه "
                "رتبه‌های Test متصل به نسخهٔ منتخب و دامنهٔ قرارداد لازم است."
            )

        selected_metrics.append({
            "metric": metric,
            "status": "BLOCKED",
            "reason_fa": reason,
            "frozen_policy": row["frozen_policy"],
            "historical_status": row["historical_status"],
            "selected_version_value": None
        })

    offline_passes = [
        row for row in preservation["scenario_outcomes"]
        if row["status"] == "PASS"
        and row.get("scope") == "OFFLINE_HISTORICAL_STATE_PRESERVATION"
    ]
    blocked = compatibility["selected_version_scenario_outcomes"]
    offline_ids = {row["regression_spec_id"] for row in offline_passes}
    blocked_ids = {row["regression_spec_id"] for row in blocked}
    original_ids = {
        row["regression_spec_id"] for row in preservation["scenario_outcomes"]
    }
    need(
        len(offline_passes) == len(offline_ids) == 26
        and len(blocked) == len(blocked_ids) == 296
        and not offline_ids.intersection(blocked_ids)
        and offline_ids | blocked_ids == original_ids
        and all(row["status"] == "BLOCKED" for row in blocked),
        "REGRESSION_ACCOUNTING_CONFLICT"
    )

    result = {
        "schema_version": "NB22E_V3_C033_R11_V1",
        "status": "R11_GATE_DECISION_RECORDED_WITH_OPEN_PREREQUISITES",
        "roadmap_action": "R11",
        "input_receipts": receipts,
        "selected_variant": selected,
        "selected_version_gate": {
            "status": "BLOCKED",
            "basis": "CURRENT_BOUND_EVIDENCE_AGAINST_FROZEN_REQUIREMENTS",
            "metrics": selected_metrics,
            "blocked_metric_count": len(selected_metrics),
            "release_accepted": False
        },
        "historical_gate": historical,
        "regression": {
            "historical_offline_checks_passed": len(offline_passes),
            "historical_offline_results": offline_passes,
            "selected_version_scenarios_blocked": len(blocked),
            "selected_version_blockers": blocked,
            "product_regression_tests_executed": 0
        },
        "subtasks": {
            "R11.1": "DONE_HISTORICAL_BLOCKERS_ACCOUNTED",
            "R11.2": "OPEN_SELECTED_VERSION_METRIC_EVIDENCE_REQUIRED",
            "R11.3": "OPEN_SELECTED_VERSION_REGRESSION_INPUTS_REQUIRED",
            "R11.4": "DONE_PROVENANCE_SCOPES_SEPARATED",
            "R11.5": "DONE_BLOCKED_DECISION_RECORDED_FOR_BOUND_EVIDENCE"
        },
        "all_R11_items_complete": False,
        "advance_to_next_roadmap_action": False,
        "next_required_evidence": [
            "رتبه‌های Test نسخهٔ منتخب روی همان Pool50 و نگاشت مرجع.",
            "پاسخ، شواهد و ارزیابی هم‌نسخه برای رفتار پاسخ و استناد.",
            "رسید Runtime و مخرج معتبر برای سنجه‌های عملیاتی."
        ]
    }

    payload = (
        json.dumps(result, ensure_ascii=False, sort_keys=True,
                   indent=2, allow_nan=False) + "\n"
    ).encode("utf-8")
    digest = hashlib.sha256(payload).hexdigest()
    output = directory / ("C033.R11_gate_disposition." + digest + ".json")
    try:
        with output.open("xb") as handle:
            handle.write(payload)
    except FileExistsError:
        need(output.read_bytes() == payload, "OUTPUT_CONTENT_CONFLICT")

    saved = output.read_bytes()
    need(
        hashlib.sha256(saved).hexdigest() == digest
        and json.loads(saved) == result,
        "OUTPUT_RELOAD_CONFLICT"
    )

    print(json.dumps({
        "status": result["status"],
        "selected_variant": selected,
        "selected_version_gate": result["selected_version_gate"]["status"],
        "historical_gate": historical["overall_status"],
        "historical_offline_checks_passed": len(offline_passes),
        "selected_version_scenarios_blocked": len(blocked),
        "subtasks": result["subtasks"],
        "all_R11_items_complete": False,
        "output": {
            "path": str(output.relative_to(root)),
            "file_bytes_sha256": digest,
            "reload_validation": "PASS"
        }
    }, ensure_ascii=False, indent=2))
    return result


V3_C033 = run_v3_c033()
````

### کار یازده — آزمون سازگاری char_rrf با Pool50 تاریخی

قرارداد ۲۶۰ سناریوی بازیابی، Pool50 و نگاشت مرجع ثابت می‌خواهد.

جست‌وجوی کاراکتری طبق تنظیمات فریزشده، برای ۲۲۶ سؤال
روی corpus پین‌شده اجرا می‌شود. خروجی هر سؤال با اعضای
Pool50 تاریخی تطبیق داده می‌شود.

اگر کاندیدای بیرونی وجود داشته باشد، ناسازگاری قرارداد ثبت می‌شود.
اگر کاندیدای بیرونی وجود نداشته باشد، احراز رتبهٔ پایهٔ
هم‌نسخه همچنان برای اجرای کامل Regression لازم است.

این اجرا آزمون سازگاری Pool است. نتیجهٔ کیفیت محصول
و وضعیت پذیرش به شواهد اجرای هم‌نسخه وابسته‌اند.

</div>

````python
# C034 — R11: اجرای جست‌وجوی کاراکتری فریز‌شده و آزمون Pool50
# وابسته به imports سل اول

def run_v3_c034():
    required = (
        "Path", "os", "json", "hashlib", "math",
        "re", "unicodedata", "collections"
    )
    missing = [name for name in required if name not in globals()]
    if missing:
        raise RuntimeError("ایمپورت لازم در سل اول: " + ", ".join(missing))

    root = Path(os.environ.get(
        "RAG_FINANCE_ROOT",
        "/Users/macbookpro/AMIR_DATA/00_WORKS/01_PROJECTS/"
        "RAG-projects/RAG_finance"
    )).expanduser().resolve()
    out_dir = root / "reports/phase2/quality_improvement/NB22E.lab.v3"
    receipts = {}

    def need(condition, message):
        if not condition:
            raise ValueError(message)

    def sha(raw):
        return hashlib.sha256(raw).hexdigest()

    def read_bound(binding, jsonl=False):
        relative = binding["path"]
        path = (root / relative).resolve()
        need(path.is_relative_to(root), "PATH_OUTSIDE_PROJECT: " + relative)
        raw = path.read_bytes()
        digest = binding["file_bytes_sha256"]
        need(sha(raw) == digest, "HASH_CONFLICT: " + relative)
        receipts[relative] = digest
        if jsonl:
            return [json.loads(line) for line in raw.splitlines() if line.strip()]
        return json.loads(raw)

    def read_v3(stem, digest):
        path = out_dir / (stem + "." + digest + ".json")
        return read_bound({
            "path": str(path.relative_to(root)),
            "file_bytes_sha256": digest
        })

    frozen = read_v3(
        "C024.R10_selected_evidence_freeze",
        "b807237f302b824c5ad943b5e3adb68a79ebd5c29c51c30edc488b88deec82e8"
    )
    previous = read_v3(
        "C033.R11_gate_disposition",
        "e62892a30721c5668efaaf23b5decdf8bd683cc05fcbabc463a40fe1f23a5b53"
    )
    need(
        frozen["selected_method"]["variant"] == previous["selected_variant"]
        == "char_rrf",
        "SELECTED_VARIANT_CONFLICT"
    )
    selected = frozen["selected_method"]
    config = selected["configuration"]
    variant_specs = [
        spec for spec in config["ranking_specs"] if spec["name"] == "char_rrf"
    ]
    need(len(variant_specs) == 1 and variant_specs[0]["channels"] == ["char"],
         "FROZEN_CHAR_RRF_RULE_CONFLICT")
    need(0 < config["incumbent_rank_weight"] < 1, "RRF_WEIGHT_CONFLICT")

    scenarios = [
        row for row in previous["regression"]["selected_version_blockers"]
        if row["kind"] == "RETRIEVAL"
    ]
    wanted = {row["candidate_id"] for row in scenarios}
    need(len(scenarios) == 260 and len(wanted) == 226, "SCENARIO_COUNT_CONFLICT")

    bindings = frozen["evaluation_lock"]["data_bindings"]
    chunks = read_bound(bindings["chunks"], jsonl=True)
    golden = read_bound(bindings["golden"], jsonl=True)
    chunk_ids = {row["chunk_id"] for row in chunks}
    question_by_id = {
        row["candidate_id"]: row["evaluation_content"]["question"]
        for row in golden if row["candidate_id"] in wanted
    }
    need(len(chunk_ids) == len(chunks), "DUPLICATE_CORPUS_CHUNK")
    need(set(question_by_id) == wanted, "QUESTION_POPULATION_CONFLICT")

    # رسید فایل تاریخی از گزارش مستقل بازتولید خوانده می‌شود.
    c056_sha = "ff6d6e136043929e0156e43f6ea75e65bd64f40ab7bcdafef90a4ba40901de22"
    verification = read_bound({
        "path": (
            "reports/phase2/release/NB22.C056.retrieval_verification/"
            "C056.independent_retrieval_metric_verification." + c056_sha + ".json"
        ),
        "file_bytes_sha256": c056_sha
    })
    historical_path = "data/phase2/retrieval/reranker_v1/reranked_candidates_v1.jsonl"
    rank_bindings = [
        receipt for receipt in verification["input_file_receipts"].values()
        if isinstance(receipt, dict) and receipt.get("path") == historical_path
    ]
    need(len(rank_bindings) == 1, "HISTORICAL_POOL50_RECEIPT_NOT_UNIQUE")
    rank_binding = rank_bindings[0]
    expected_hashes = {
        rank_binding[key] for key in ("file_bytes_sha256", "sha256")
        if key in rank_binding
    }
    need(len(expected_hashes) == 1, "HISTORICAL_RANK_HASH_NOT_UNIQUE")
    historical_rows = read_bound({
        "path": historical_path,
        "file_bytes_sha256": expected_hashes.pop()
    }, jsonl=True)

    pools = {}
    for row in historical_rows:
        candidate_id = row["candidate_id"]
        if candidate_id not in wanted:
            continue
        need(candidate_id not in pools, "DUPLICATE_POOL50: " + candidate_id)
        need(row["split"] == "test", "HISTORICAL_SPLIT_CONFLICT: " + candidate_id)
        candidates = row["candidates"]
        need(
            len(candidates) == 50
            and [item["rank"] for item in candidates] == list(range(1, 51)),
            "HISTORICAL_POOL50_SCHEMA_CONFLICT: " + candidate_id
        )
        ids = {item["chunk_id"] for item in candidates}
        need(len(ids) == 50 and ids.issubset(chunk_ids),
             "HISTORICAL_POOL50_ID_CONFLICT: " + candidate_id)
        pools[candidate_id] = ids
    need(set(pools) == wanted, "HISTORICAL_POOL50_POPULATION_CONFLICT")

    # همان تحلیل‌گر و BM25 کاراکتریِ تولیدکنندهٔ NB22D.
    translation = str.maketrans(
        "يك٠١٢٣٤٥٦٧٨٩۰۱۲۳۴۵۶۷۸۹", "یک01234567890123456789"
    )

    def char_tokens(text):
        text = unicodedata.normalize("NFKC", text).lower().translate(translation)
        text = "".join(
            character for character in text
            if not unicodedata.combining(character) and character != "ـ"
        ).replace("\u200c", " ")
        words = re.findall(r"[^\W_]+", text, flags=re.UNICODE)
        return [
            word[index:index + 3]
            for word in words
            for index in range(len(word) - 2)
        ]

    ordered_chunks = sorted(chunks, key=lambda row: row["chunk_id"])
    corpus_ids = [row["chunk_id"] for row in ordered_chunks]
    postings = collections.defaultdict(list)
    lengths = []
    for position, row in enumerate(ordered_chunks):
        frequencies = collections.Counter(char_tokens(row["content"]))
        lengths.append(sum(frequencies.values()))
        for token, frequency in frequencies.items():
            postings[token].append((position, frequency))

    corpus_size = len(corpus_ids)
    average_length = sum(lengths) / corpus_size
    need(average_length > 0, "EMPTY_CHAR_INDEX")
    k1 = config["bm25"]["k1"]
    b = config["bm25"]["b"]
    max_df = config["char_max_df_fraction"]

    def lexical_order(question):
        scores = {}
        for token in sorted(set(char_tokens(question))):
            posting = postings.get(token, ())
            df = len(posting)
            if df == 0 or df / corpus_size > max_df:
                continue
            idf = math.log(1 + (corpus_size - df + .5) / (df + .5))
            for position, frequency in posting:
                norm = 1 - b + b * lengths[position] / average_length
                contribution = idf * frequency * (k1 + 1) / (
                    frequency + k1 * norm
                )
                scores[position] = scores.get(position, 0.0) + contribution
        return [
            corpus_ids[position]
            for position in sorted(
                scores, key=lambda position: (-scores[position], corpus_ids[position])
            )[:800]
            if scores[position] > 0
        ]

    question_results = {}
    for candidate_id in sorted(wanted):
        order = lexical_order(question_by_id[candidate_id])
        outside = [chunk_id for chunk_id in order if chunk_id not in pools[candidate_id]]
        question_results[candidate_id] = {
            "candidate_id": candidate_id,
            "question_sha256": sha(question_by_id[candidate_id].encode("utf-8")),
            "historical_pool50_chunk_ids": sorted(pools[candidate_id]),
            "lexical_candidate_count": len(order),
            "outside_pool50_count": len(outside),
            "outside_pool50_examples": outside[:5],
            "compatibility_status": (
                "INCOMPATIBLE_FROZEN_FUSION_ADDS_OUTSIDE_POOL50"
                if outside else "POOL_MEMBERSHIP_COMPATIBLE_INCUMBENT_BINDING_REQUIRED"
            )
        }

    scenario_results = [{
        "regression_spec_id": row["regression_spec_id"],
        "candidate_id": row["candidate_id"],
        "pool_compatibility": question_results[row["candidate_id"]]["compatibility_status"],
        "selected_version_regression_status": "BLOCKED"
    } for row in scenarios]

    counts = dict(collections.Counter(
        row["pool_compatibility"] for row in scenario_results
    ))
    result = {
        "schema_version": "NB22E_V3_C034_R11_V1",
        "status": "R11_FIXED_POOL_COMPATIBILITY_TEST_EXECUTED",
        "roadmap_action": "R11",
        "selected_variant": "char_rrf",
        "input_receipts": receipts,
        "questions_scored_locally": len(question_results),
        "scenario_compatibility_counts": counts,
        "question_results": list(question_results.values()),
        "scenario_results": scenario_results,
        "product_regression_tests_executed": 0,
        "new_api_calls": 0,
        "new_model_pairs": 0,
        "selected_version_gate": previous["selected_version_gate"]["status"],
        "all_R11_items_complete": False
    }

    payload = (
        json.dumps(result, ensure_ascii=False, sort_keys=True,
                   indent=2, allow_nan=False) + "\n"
    ).encode("utf-8")
    digest = sha(payload)
    output = out_dir / ("C034.R11_fixed_pool_compatibility." + digest + ".json")
    try:
        with output.open("xb") as handle:
            handle.write(payload)
    except FileExistsError:
        need(output.read_bytes() == payload, "OUTPUT_CONTENT_CONFLICT")

    saved = output.read_bytes()
    need(sha(saved) == digest and json.loads(saved) == result,
         "OUTPUT_RELOAD_CONFLICT")
    print(json.dumps({
        "status": result["status"],
        "questions_scored_locally": len(question_results),
        "scenario_compatibility_counts": counts,
        "product_regression_tests_executed": 0,
        "output": {
            "path": str(output.relative_to(root)),
            "file_bytes_sha256": digest,
            "reload_validation": "PASS"
        }
    }, ensure_ascii=False, indent=2))
    return result


V3_C034 = run_v3_c034()
````

### کار یازده — ثبت نتیجهٔ آزمون سازگاری Pool50

جست‌وجوی کاراکتری روی ۲۲۶ سؤال Test تاریخی انجام شد.
در هر سؤال، دست‌کم یک کاندیدا بیرون از Pool50 تاریخی بود.

۳۱ سناریوی رفتار پاسخ و پنج سناریوی استناد به خروجی

وضعیت Gate نسخهٔ منتخب BLOCKED است.
ارزیابی بازیابی با Pool گسترده‌تر به قرارداد آزمون جداگانه نیاز دارد.

````python
# C035 — R11: اتصال نتیجهٔ Pool50 به دفتر سناریوها
# وابسته به imports سل اول

def run_v3_c035():
    missing = [
        name for name in ("Path", "os", "json", "hashlib")
        if name not in globals()
    ]
    if missing:
        raise RuntimeError("ابتدا سل imports را اجرا کنید: " + ", ".join(missing))

    def need(condition, message):
        if not condition:
            raise ValueError(message)

    def sha(raw):
        return hashlib.sha256(raw).hexdigest()

    root = Path(os.environ.get(
        "RAG_FINANCE_ROOT",
        "/Users/macbookpro/AMIR_DATA/00_WORKS/01_PROJECTS/"
        "RAG-projects/RAG_finance"
    )).expanduser().resolve()
    directory = root / "reports/phase2/quality_improvement/NB22E.lab.v3"
    receipts = {}

    def read_pinned(stem, digest):
        path = directory / (stem + "." + digest + ".json")
        raw = path.read_bytes()
        need(sha(raw) == digest, "HASH_CONFLICT: " + stem)
        receipts[stem.split(".")[0]] = {
            "path": str(path.relative_to(root)),
            "file_bytes_sha256": digest
        }
        return json.loads(raw)

    previous = read_pinned(
        "C033.R11_gate_disposition",
        "e62892a30721c5668efaaf23b5decdf8bd683cc05fcbabc463a40fe1f23a5b53"
    )
    compatibility = read_pinned(
        "C034.R11_fixed_pool_compatibility",
        "73fa7128f709137d002b577c60b780e35124e61950137ea1704af0955492010b"
    )

    need(
        previous["selected_variant"] == compatibility["selected_variant"]
        == "char_rrf",
        "SELECTED_VARIANT_CONFLICT"
    )
    need(
        compatibility["input_receipts"][receipts["C033"]["path"]]
        == receipts["C033"]["file_bytes_sha256"],
        "C034_PARENT_RECEIPT_CONFLICT"
    )
    need(
        previous["selected_version_gate"]["status"] == "BLOCKED"
        and compatibility["selected_version_gate"] == "BLOCKED",
        "GATE_STATE_CONFLICT"
    )

    regression = previous["regression"]
    original_blockers = regression["selected_version_blockers"]
    original_ids = {row["regression_spec_id"] for row in original_blockers}
    retrieval_ids = {
        row["regression_spec_id"] for row in original_blockers
        if row["kind"] == "RETRIEVAL"
    }
    need(
        len(original_blockers) == len(original_ids) == 296
        and len(retrieval_ids) == 260
        and regression["historical_offline_checks_passed"] == 26,
        "SCENARIO_POPULATION_CONFLICT"
    )

    tested = compatibility["scenario_results"]
    tested_by_id = {row["regression_spec_id"]: row for row in tested}
    question_rows = compatibility["question_results"]
    questions = {row["candidate_id"]: row for row in question_rows}

    incompatibility = "INCOMPATIBLE_FROZEN_FUSION_ADDS_OUTSIDE_POOL50"
    need(
        len(tested) == len(tested_by_id) == 260
        and set(tested_by_id) == retrieval_ids
        and len(question_rows) == len(questions) == 226
        and compatibility["questions_scored_locally"] == 226,
        "COMPATIBILITY_POPULATION_CONFLICT"
    )
    need(
        {row["candidate_id"] for row in tested} == set(questions),
        "QUESTION_BINDING_CONFLICT"
    )

    updated = []
    counts = {"RETRIEVAL": 0, "ANSWER_DECISION": 0, "CITATION_WITNESS": 0}

    for old_row in original_blockers:
        spec_id = old_row["regression_spec_id"]
        kind = old_row["kind"]
        need(kind in counts, spec_id + ": UNKNOWN_SCENARIO_KIND")
        need(old_row["status"] == "BLOCKED", spec_id + ": PRIOR_STATE_CONFLICT")
        counts[kind] += 1
        row = dict(old_row)

        if kind == "RETRIEVAL":
            evidence = tested_by_id[spec_id]
            candidate_id = row["candidate_id"]
            question = questions[candidate_id]
            need(
                evidence["candidate_id"] == candidate_id
                and evidence["pool_compatibility"] == incompatibility
                and evidence["selected_version_regression_status"] == "BLOCKED"
                and question["compatibility_status"] == incompatibility
                and question["outside_pool50_count"] > 0,
                spec_id + ": POOL_COMPATIBILITY_EVIDENCE_CONFLICT"
            )
            row.update({
                "previous_blocker": old_row["blocker"],
                "blocker": "FROZEN_CHAR_RRF_INCOMPATIBLE_WITH_FIXED_POOL50",
                "required_input": (
                    "A separately versioned evaluation contract for the "
                    "expanded candidate pool, with a defined reference, "
                    "baseline, and comparison scope."
                ),
                "compatibility_evidence": {
                    "source": receipts["C034"],
                    "question_sha256": question["question_sha256"],
                    "outside_pool50_count": question["outside_pool50_count"],
                    "outside_pool50_examples": question["outside_pool50_examples"]
                }
            })

        updated.append(row)

    need(
        counts == {"RETRIEVAL": 260, "ANSWER_DECISION": 31, "CITATION_WITNESS": 5},
        "FINAL_SCENARIO_COUNTS_CONFLICT"
    )

    result = {
        "schema_version": "NB22E_V3_C035_R11_V1",
        "status": "R11_FIXED_POOL_BLOCKER_BOUND_TO_REGRESSION_LEDGER",
        "roadmap_action": "R11",
        "input_receipts": receipts,
        "selected_variant": previous["selected_variant"],
        "selected_version_gate": previous["selected_version_gate"],
        "historical_gate": previous["historical_gate"],
        "scenario_counts": {
            "historical_offline_preservation_pass": 26,
            "retrieval_blocked_by_fixed_pool_contract": 260,
            "answer_decision_blocked_pending_version_matched_output": 31,
            "citation_blocked_pending_version_matched_evidence": 5
        },
        "historical_offline_results": regression["historical_offline_results"],
        "selected_version_blockers": updated,
        "pool_compatibility_questions_scored_in_C034": 226,
        "product_regression_tests_executed": 0,
        "subtasks": previous["subtasks"],
        "all_R11_items_complete": False,
        "advance_to_next_roadmap_action": False,
        "next_work": {
            "retrieval": "DEFINE_SEPARATE_EXPANDED_POOL_EVALUATION_CONTRACT",
            "answer_and_citation": "BIND_SELECTED_VERSION_RESPONSES_AND_EVIDENCE"
        }
    }

    payload = (
        json.dumps(result, ensure_ascii=False, sort_keys=True,
                   indent=2, allow_nan=False) + "\n"
    ).encode("utf-8")
    digest = sha(payload)
    output = directory / ("C035.R11_fixed_pool_disposition." + digest + ".json")

    try:
        with output.open("xb") as handle:
            handle.write(payload)
    except FileExistsError:
        need(output.read_bytes() == payload, "OUTPUT_CONTENT_CONFLICT")

    saved = output.read_bytes()
    need(
        sha(saved) == digest and json.loads(saved) == result,
        "OUTPUT_RELOAD_CONFLICT"
    )
    print(json.dumps({
        "status": result["status"],
        "scenario_counts": result["scenario_counts"],
        "selected_version_gate": result["selected_version_gate"]["status"],
        "historical_gate": result["historical_gate"]["overall_status"],
        "product_regression_tests_executed": 0,
        "all_R11_items_complete": False,
        "output": {
            "path": str(output.relative_to(root)),
            "file_bytes_sha256": digest,
            "reload_validation": "PASS"
        }
    }, ensure_ascii=False, indent=2))
    return result


V3_C035 = run_v3_c035()
````

### کار یازده — اجرای مقایسهٔ تشخیصی با Pool گسترده

برای هر یک از ۲۲۶ سؤال، ترتیب تاریخی Pool50 با خروجی ترکیب
همان ترتیب و کانال کاراکتری مقایسه می‌شود.

````python
# C036 — R11: مقایسهٔ تشخیصی Pool گسترده و ثبت ورودی پاسخ/استناد
# وابسته به imports سل اول

def run_v3_c036():
    required = (
        "Path", "os", "json", "hashlib", "math", "re",
        "unicodedata", "collections", "ast"
    )
    missing = [name for name in required if name not in globals()]
    if missing:
        raise RuntimeError("ایمپورت لازم در سل اول: " + ", ".join(missing))

    root = Path(os.environ.get(
        "RAG_FINANCE_ROOT",
        "/Users/macbookpro/AMIR_DATA/00_WORKS/01_PROJECTS/"
        "RAG-projects/RAG_finance"
    )).expanduser().resolve()
    directory = root / "reports/phase2/quality_improvement/NB22E.lab.v3"
    receipts = {}

    def need(condition, message):
        if not condition:
            raise ValueError(message)

    def sha(raw):
        return hashlib.sha256(raw).hexdigest()

    def canonical(value):
        return json.dumps(
            value, ensure_ascii=False, sort_keys=True,
            separators=(",", ":"), allow_nan=False
        ).encode("utf-8")

    def read_bound(relative, digest, jsonl=False):
        path = (root / relative).resolve()
        need(path.is_relative_to(root), "PATH_OUTSIDE_PROJECT: " + relative)
        raw = path.read_bytes()
        need(sha(raw) == digest, "HASH_CONFLICT: " + relative)
        receipts[relative] = digest
        if jsonl:
            return [json.loads(line) for line in raw.splitlines() if line.strip()]
        return json.loads(raw)

    def read_v3(stem, digest):
        path = directory / (stem + "." + digest + ".json")
        return read_bound(str(path.relative_to(root)), digest)

    def save(stem, value):
        payload = (
            json.dumps(value, ensure_ascii=False, sort_keys=True,
                       indent=2, allow_nan=False) + "\n"
        ).encode("utf-8")
        digest = sha(payload)
        path = directory / (stem + "." + digest + ".json")
        try:
            with path.open("xb") as handle:
                handle.write(payload)
        except FileExistsError:
            need(path.read_bytes() == payload, "OUTPUT_CONTENT_CONFLICT")
        saved = path.read_bytes()
        need(sha(saved) == digest and json.loads(saved) == value,
             "OUTPUT_RELOAD_CONFLICT")
        return {
            "path": str(path.relative_to(root)),
            "file_bytes_sha256": digest,
            "reload_validation": "PASS"
        }

    frozen = read_v3(
        "C024.R10_selected_evidence_freeze",
        "b807237f302b824c5ad943b5e3adb68a79ebd5c29c51c30edc488b88deec82e8"
    )
    definitions = read_v3(
        "C028.R11_original_regression_definitions",
        "9cd26651bf9794b4ef993d899bdf844c113930e96c43a0a6feffbca540565c08"
    )
    compatibility = read_v3(
        "C034.R11_fixed_pool_compatibility",
        "73fa7128f709137d002b577c60b780e35124e61950137ea1704af0955492010b"
    )
    previous = read_v3(
        "C035.R11_fixed_pool_disposition",
        "e81b6b8bc613e6b1158558dd47b2db06ff943402e2ccb1a3c42b04966e744e1d"
    )

    selected = frozen["selected_method"]
    need(
        selected["variant"] == previous["selected_variant"]
        == compatibility["selected_variant"] == "char_rrf",
        "SELECTED_VARIANT_CONFLICT"
    )

    code_cells = selected["producer_code_snapshot"]
    need(
        sha(canonical(code_cells)) == selected["producer_code_sha256"],
        "FROZEN_PRODUCER_CODE_HASH_CONFLICT"
    )
    config = selected["configuration"]
    need(
        sha(canonical(config)) == selected["configuration_canonical_sha256"],
        "FROZEN_CONFIG_HASH_CONFLICT"
    )
    variants = [
        item for item in config["ranking_specs"] if item["name"] == "char_rrf"
    ]
    need(len(variants) == 1 and variants[0]["channels"] == ["char"],
         "FROZEN_VARIANT_CONFIGURATION_CONFLICT")

    # فقط توابع خوانده‌شدهٔ جست‌وجو و ترکیب از کد پین‌شده بار می‌شوند.
    function_names = {
        "normalized_words", "char_tokens", "build_index",
        "lexical_search", "lexical_fusion"
    }
    functions = {}
    for text in code_cells:
        for node in ast.parse(text).body:
            if isinstance(node, ast.FunctionDef) and node.name in function_names:
                need(node.name not in functions and not node.decorator_list,
                     "FROZEN_FUNCTION_IDENTITY_CONFLICT: " + node.name)
                functions[node.name] = node
    need(set(functions) == function_names, "FROZEN_FUNCTIONS_MISSING")

    namespace = {
        "CONFIG": config,
        "collections": collections,
        "math": math,
        "re": re,
        "unicodedata": unicodedata,
        "require": need
    }
    module = ast.Module(body=list(functions.values()), type_ignores=[])
    exec(compile(module, "<pinned-NB22D-retrieval-functions>", "exec"), namespace)

    documents = definitions["original_regression_sources"]
    catalogue = {
        row["issue_id"]: row
        for row in documents["final_issue_catalogue"]["content"]
    }
    specs = {
        row["regression_spec_id"]: row
        for row in documents["original_regression_definitions"]["content"]
    }
    pool_records = {
        row["candidate_id"]: row for row in compatibility["question_results"]
    }

    references = {}
    scenario_ids = {}
    response_requirements = []

    for row in previous["selected_version_blockers"]:
        spec = specs[row["regression_spec_id"]]
        candidate_id = row["candidate_id"]
        need(spec["candidate_id"] == candidate_id, "SCENARIO_IDENTITY_CONFLICT")
        if row["kind"] == "RETRIEVAL":
            issue = catalogue[spec["issue_id"]]
            gold = set(issue["witness"]["relevant_chunk_ids"])
            need(bool(gold), "EMPTY_HISTORICAL_REFERENCE: " + candidate_id)
            need(
                candidate_id not in references or references[candidate_id] == gold,
                "REFERENCE_CONFLICT: " + candidate_id
            )
            references[candidate_id] = gold
            scenario_ids.setdefault(candidate_id, []).append(spec["regression_spec_id"])
        else:
            response_requirements.append({
                "regression_spec_id": spec["regression_spec_id"],
                "candidate_id": candidate_id,
                "kind": row["kind"],
                "assertion": spec["assertion"],
                "historical_response_sha256":
                    spec["frozen_response_document_sha256"],
                "historical_evidence": row["bound_evidence"],
                "required_selected_version_input": row["required_input"],
                "status": "BLOCKED"
            })

    need(
        len(references) == len(pool_records) == 226
        and set(references) == set(pool_records)
        and sum(map(len, scenario_ids.values())) == 260
        and len(response_requirements) == 36,
        "R11_POPULATION_CONFLICT"
    )

    historical_path = "data/phase2/retrieval/reranker_v1/reranked_candidates_v1.jsonl"
    rank_digest = compatibility["input_receipts"][historical_path]
    rank_rows = read_bound(historical_path, rank_digest, jsonl=True)
    baselines = {}
    for row in rank_rows:
        candidate_id = row["candidate_id"]
        if candidate_id not in references:
            continue
        need(candidate_id not in baselines and row["split"] == "test",
             "BASELINE_IDENTITY_CONFLICT: " + candidate_id)
        candidates = row["candidates"]
        need([item["rank"] for item in candidates] == list(range(1, 51)),
             "BASELINE_RANKS_CONFLICT: " + candidate_id)
        order = [item["chunk_id"] for item in candidates]
        need(
            len(set(order)) == 50
            and set(order) == set(pool_records[candidate_id]["historical_pool50_chunk_ids"]),
            "POOL50_IDENTITY_CONFLICT: " + candidate_id
        )
        baselines[candidate_id] = order
    need(set(baselines) == set(references), "BASELINE_QUERY_SET_CONFLICT")

    bindings = frozen["evaluation_lock"]["data_bindings"]
    chunk_rows = read_bound(
        bindings["chunks"]["path"], bindings["chunks"]["file_bytes_sha256"],
        jsonl=True
    )
    golden_rows = read_bound(
        bindings["golden"]["path"], bindings["golden"]["file_bytes_sha256"],
        jsonl=True
    )
    chunks = {row["chunk_id"]: row for row in chunk_rows}
    questions = {
        row["candidate_id"]: row["evaluation_content"]["question"]
        for row in golden_rows if row["candidate_id"] in references
    }
    need(len(chunks) == len(chunk_rows), "DUPLICATE_CORPUS_CHUNK")
    need(set(questions) == set(references), "QUESTION_SET_CONFLICT")
    for candidate_id, question in questions.items():
        need(
            sha(question.encode("utf-8")) == pool_records[candidate_id]["question_sha256"]
            and references[candidate_id].issubset(chunks),
            "QUESTION_OR_REFERENCE_BINDING_CONFLICT: " + candidate_id
        )

    plan = {
        "version": "R11_EXPANDED_POOL_HISTORICAL_BASELINE_DIAGNOSTIC_V1",
        "scope": "PREVIOUSLY_EXPOSED_HISTORICAL_TEST_FAILURE_SUBSET",
        "candidate_ids": sorted(references),
        "cutoff": 10,
        "aggregation": "MACRO_PER_DISTINCT_QUERY",
        "baseline": "PINNED_HISTORICAL_POOL50_ORDER",
        "candidate": "FROZEN_CHAR_FUSION_APPLIED_TO_HISTORICAL_POOL50_INCUMBENT",
        "reference": "C078_HISTORICAL_REFERENCE_CHUNK_IDS",
        "producer_code_sha256": selected["producer_code_sha256"],
        "configuration_sha256": selected["configuration_canonical_sha256"],
        "input_receipts": dict(receipts)
    }
    plan_output = save("C036.R11_expanded_pool_plan", plan)

    def measure(order, gold):
        hits = [int(chunk_id in gold) for chunk_id in order[:10]]
        ideal = math.fsum(
            1 / math.log2(i + 2) for i in range(min(10, len(gold)))
        )
        return {
            "recall_at_10": sum(hits) / len(gold),
            "ndcg_at_10": math.fsum(
                hit / math.log2(i + 2) for i, hit in enumerate(hits)
            ) / ideal,
            "hit_at_10": float(any(hits))
        }

    index = namespace["build_index"](chunks, "char")
    results = []

    for candidate_id in sorted(references):
        historical_order = baselines[candidate_id]
        lexical_order = namespace["lexical_search"](
            index, questions[candidate_id], "char"
        )
        outside_count = len(set(lexical_order) - set(historical_order))
        need(
            outside_count == pool_records[candidate_id]["outside_pool50_count"],
            "C034_LEXICAL_COUNT_CONFLICT: " + candidate_id
        )
        expanded_order = namespace["lexical_fusion"](
            historical_order, {"char": lexical_order}, variants[0]
        )
        need(
            len(expanded_order) == len(set(expanded_order))
            and set(expanded_order) == set(historical_order) | set(lexical_order),
            "FUSION_MEMBERSHIP_CONFLICT: " + candidate_id
        )
        gold = references[candidate_id]
        results.append({
            "candidate_id": candidate_id,
            "original_regression_spec_ids": sorted(scenario_ids[candidate_id]),
            "historical_reference_chunk_ids": sorted(gold),
            "baseline_top10": historical_order[:10],
            "expanded_top10": expanded_order[:10],
            "expanded_candidate_count": len(expanded_order),
            "baseline_metrics": measure(historical_order, gold),
            "expanded_metrics": measure(expanded_order, gold)
        })

    metrics = ("recall_at_10", "ndcg_at_10", "hit_at_10")
    macro = {
        arm: {
            metric: math.fsum(row[arm][metric] for row in results) / len(results)
            for metric in metrics
        }
        for arm in ("baseline_metrics", "expanded_metrics")
    }
    delta = {
        metric: macro["expanded_metrics"][metric] - macro["baseline_metrics"][metric]
        for metric in metrics
    }
    hit_transitions = {
        "miss_to_hit": sum(
            row["baseline_metrics"]["hit_at_10"] == 0
            and row["expanded_metrics"]["hit_at_10"] == 1 for row in results
        ),
        "hit_to_miss": sum(
            row["baseline_metrics"]["hit_at_10"] == 1
            and row["expanded_metrics"]["hit_at_10"] == 0 for row in results
        )
    }

    result = {
        "schema_version": "NB22E_V3_C036_R11_V1",
        "status": "R11_EXPANDED_POOL_DIAGNOSTIC_EXECUTED",
        "roadmap_action": "R11",
        "plan_receipt": plan_output,
        "input_receipts": receipts,
        "questions_evaluated": len(results),
        "macro": macro,
        "expanded_minus_baseline": delta,
        "hit_transitions": hit_transitions,
        "per_query": results,
        "original_fixed_pool_scenarios_blocked": 260,
        "response_and_citation_requirements": response_requirements,
        "product_regression_tests_executed": 0,
        "new_api_calls": 0,
        "new_model_pairs": 0,
        "selected_version_gate": previous["selected_version_gate"]["status"],
        "all_R11_items_complete": False,
        "advance_to_next_roadmap_action": False
    }
    output = save("C036.R11_expanded_pool_diagnostic", result)

    print(json.dumps({
        "status": result["status"],
        "questions_evaluated": len(results),
        "macro": macro,
        "expanded_minus_baseline": delta,
        "hit_transitions": hit_transitions,
        "response_and_citation_scenarios_pending": len(response_requirements),
        "selected_version_gate": result["selected_version_gate"],
        "output": output
    }, ensure_ascii=False, indent=2))
    return result


V3_C036 = run_v3_c036()
````

````python
# C037 — R11: ساخت ورودی اجرایی برای ۳۶ سناریوی پاسخ و استناد
# وابسته به imports سل اول

def run_v3_c037():
    required = ("Path", "os", "json", "hashlib")
    missing = [name for name in required if name not in globals()]
    if missing:
        raise RuntimeError("ابتدا سل imports را اجرا کنید: " + ", ".join(missing))

    root = Path(os.environ.get(
        "RAG_FINANCE_ROOT",
        "/Users/macbookpro/AMIR_DATA/00_WORKS/01_PROJECTS/"
        "RAG-projects/RAG_finance"
    )).expanduser().resolve()
    directory = root / "reports/phase2/quality_improvement/NB22E.lab.v3"
    receipts = {}

    def need(condition, message):
        if not condition:
            raise ValueError(message)

    def sha(raw):
        return hashlib.sha256(raw).hexdigest()

    def read_pinned(stem, digest):
        path = directory / (stem + "." + digest + ".json")
        raw = path.read_bytes()
        need(sha(raw) == digest, "HASH_CONFLICT: " + stem)
        receipts[stem.split(".")[0]] = {
            "path": str(path.relative_to(root)),
            "file_bytes_sha256": digest
        }
        return json.loads(raw)

    definitions = read_pinned(
        "C028.R11_original_regression_definitions",
        "9cd26651bf9794b4ef993d899bdf844c113930e96c43a0a6feffbca540565c08"
    )
    historical = read_pinned(
        "C031.R11_answer_citation_binding",
        "4950b86eb86e688f840f2d2a464b054e2613c13a2e2872bed137825e3da67b6a"
    )
    diagnostic = read_pinned(
        "C036.R11_expanded_pool_diagnostic",
        "69a6ffdbd965ce8a2a6176ba201445798d1604c84e73954a6a407b7411f2b7b2"
    )

    need(
        diagnostic["input_receipts"][receipts["C028"]["path"]]
        == receipts["C028"]["file_bytes_sha256"],
        "C036_C028_RECEIPT_CONFLICT"
    )
    need(
        historical["input_receipts"]["C028.R11_original_regression_definitions"]
        == receipts["C028"],
        "C031_C028_RECEIPT_CONFLICT"
    )
    need(
        diagnostic["selected_version_gate"] == "BLOCKED"
        and diagnostic["questions_evaluated"] == 226
        and diagnostic["original_fixed_pool_scenarios_blocked"] == 260,
        "R11_DIAGNOSTIC_STATE_CONFLICT"
    )

    def index_by(rows, key):
        indexed = {}
        for row in rows:
            identifier = row[key]
            need(identifier not in indexed, "DUPLICATE_" + key + ": " + identifier)
            indexed[identifier] = row
        return indexed

    sources = definitions["original_regression_sources"]
    specs = index_by(
        sources["original_regression_definitions"]["content"],
        "regression_spec_id"
    )
    issues = index_by(sources["final_issue_catalogue"]["content"], "issue_id")
    historical_rows = index_by(
        historical["answer_scenarios"] + historical["citation_scenarios"],
        "regression_spec_id"
    )
    pending = index_by(
        diagnostic["response_and_citation_requirements"], "regression_spec_id"
    )
    need(
        len(pending) == len(historical_rows) == 36
        and set(pending) == set(historical_rows),
        "R11_PENDING_SCENARIO_SET_CONFLICT"
    )

    tasks = []
    counts = {"ANSWER_DECISION": 0, "CITATION_WITNESS": 0}

    for spec_id, item in sorted(pending.items()):
        spec = specs[spec_id]
        issue = issues[spec["issue_id"]]
        prior = historical_rows[spec_id]
        candidate_id = item["candidate_id"]
        response_sha = spec["frozen_response_document_sha256"]
        kind = item["kind"]

        need(kind in counts, "UNEXPECTED_SCENARIO_KIND: " + spec_id)
        need(
            item["status"] == prior["selected_version_test_status"] == "BLOCKED"
            and candidate_id == spec["candidate_id"]
            == issue["candidate_id"] == prior["candidate_id"]
            and issue["regression_spec_id"] == spec_id
            and response_sha == issue["response_document_sha256"]
            == prior["historical_response_sha256"]
            == item["historical_response_sha256"],
            "SCENARIO_OR_RESPONSE_BINDING_CONFLICT: " + spec_id
        )

        task = {
            "regression_spec_id": spec_id,
            "issue_id": spec["issue_id"],
            "candidate_id": candidate_id,
            "kind": kind,
            "assertion": spec["assertion"],
            "historical_response_sha256": response_sha,
            "selected_version_execution_status": "BLOCKED",
            "required_execution_evidence": [
                "SELECTED_VARIANT_AND_RUNTIME_RECEIPT",
                "QUESTION_AND_EVIDENCE_BUNDLE_IDENTITY",
                "SELECTED_RESPONSE_DOCUMENT_AND_BYTE_HASH",
                "ASSESSMENT_BOUND_TO_SELECTED_RESPONSE_AND_SCENARIO"
            ],
            "historical_A07_response_hash_matches":
                prior["historical_A07_response_hash_matches"]
        }

        if kind == "ANSWER_DECISION":
            need(
                spec["kind"] == "SOURCE_BOUND_ANSWER_DECISION_REGRESSION",
                "ANSWER_SPEC_KIND_CONFLICT: " + spec_id
            )
            task["behavior_expectation"] = {
                "expected_behavior": spec["expected_behavior"],
                "historical_observed_behavior": spec["historical_observed_behavior"],
                "historical_required_limit_disclosed":
                    spec["required_limit_disclosed"],
                "historical_decision_assessment":
                    prior["historical_decision_assessment"]
            }
            task["required_execution_evidence"].append(
                "EXPECTED_BEHAVIOR_AND_REQUIRED_LIMIT_ASSESSED"
            )

        else:
            need(
                spec["kind"] == "CITATION_WITNESS_RECORDING_REGRESSION",
                "CITATION_SPEC_KIND_CONFLICT: " + spec_id
            )
            witness = issue["witness"]
            need(
                witness["claim_index"] == prior["claim_index"]
                and witness["evidence_id"] == prior["evidence_id"],
                "HISTORICAL_CITATION_LINK_CONFLICT: " + spec_id
            )
            evidence_text = witness["evidence_text"]
            spans = prior["historical_exact_spans"]
            for span in spans:
                start, end = span["start"], span["end"]
                need(
                    0 <= start < end <= len(evidence_text)
                    and evidence_text[start:end] == span["quote"],
                    "HISTORICAL_QUOTE_SPAN_CONFLICT: " + spec_id
                )

            support = prior["historical_model_link_support"]
            need(
                (support == "SUPPORTED" and bool(spans))
                or (support == "NOT_SUPPORTED" and not spans),
                "HISTORICAL_LINK_VERDICT_CONFLICT: " + spec_id
            )
            task["historical_link_expectation"] = {
                "claim_index": witness["claim_index"],
                "claim": witness["claim"],
                "evidence_id": witness["evidence_id"],
                "evidence_text_sha256": sha(evidence_text.encode("utf-8")),
                "model_support": support,
                "exact_spans": spans,
                "span_unit": "UNICODE_CODE_POINT"
            }
            task["required_execution_evidence"].append(
                "EXACT_SELECTED_CLAIM_LINK_AND_QUALIFIED_WITNESS_OR_EXPLICIT_VERDICT"
            )

        tasks.append(task)
        counts[kind] += 1

    need(
        counts == {"ANSWER_DECISION": 31, "CITATION_WITNESS": 5},
        "R11_TASK_COUNTS_CONFLICT"
    )

    result = {
        "schema_version": "NB22E_V3_C037_R11_V1",
        "status": "R11_RESPONSE_AND_CITATION_EXECUTION_INPUTS_SPECIFIED",
        "roadmap_action": "R11",
        "input_receipts": receipts,
        "selected_variant": historical["selected_variant"],
        "selected_version_gate": diagnostic["selected_version_gate"],
        "scenario_counts": counts,
        "distinct_candidate_count": len({task["candidate_id"] for task in tasks}),
        "selected_version_execution_requests": tasks,
        "fixed_pool_retrieval_scenarios_blocked": 260,
        "product_regression_tests_executed": 0,
        "new_api_calls": 0,
        "all_R11_items_complete": False,
        "advance_to_next_roadmap_action": False,
        "next_step": "BIND_SELECTED_RUNTIME_RESPONSES_AND_SCOPED_ASSESSMENTS"
    }

    payload = (
        json.dumps(result, ensure_ascii=False, sort_keys=True,
                   indent=2, allow_nan=False) + "\n"
    ).encode("utf-8")
    digest = sha(payload)
    output = directory / ("C037.R11_response_regression_inputs." + digest + ".json")
    try:
        with output.open("xb") as handle:
            handle.write(payload)
    except FileExistsError:
        need(output.read_bytes() == payload, "OUTPUT_CONTENT_CONFLICT")

    saved = output.read_bytes()
    need(sha(saved) == digest and json.loads(saved) == result,
         "OUTPUT_RELOAD_CONFLICT")

    print(json.dumps({
        "status": result["status"],
        "scenario_counts": counts,
        "distinct_candidate_count": result["distinct_candidate_count"],
        "selected_version_gate": result["selected_version_gate"],
        "product_regression_tests_executed": 0,
        "output": {
            "path": str(output.relative_to(root)),
            "file_bytes_sha256": digest,
            "reload_validation": "PASS"
        }
    }, ensure_ascii=False, indent=2))
    return result


V3_C037 = run_v3_c037()
````

### کار یازده — اجرای Regression پاسخ و استناد با رسید هم‌نسخه

````python
# C038 — اجرای کنترل‌های ردیفی روی رسیدهای پاسخ و ارزیابی نسخهٔ منتخب
# imports از سل اول: Path, os, json, hashlib

# مسیر نسبی و SHA-256 بستهٔ واقعی Runner را پس از تولید آن وارد کنید.
# اجرای بدون بسته، درخواست‌های دقیقِ لازم را در خروجی ثبت می‌کند.
C038_RUN_BUNDLE_PATH = None
C038_RUN_BUNDLE_SHA256 = None


def run_v3_c038(bundle_path, bundle_sha):
    root = Path(os.environ.get(
        "RAG_FINANCE_ROOT",
        "/Users/macbookpro/AMIR_DATA/00_WORKS/01_PROJECTS/"
        "RAG-projects/RAG_finance"
    )).expanduser().resolve()
    directory = root / "reports/phase2/quality_improvement/NB22E.lab.v3"
    freeze_sha = "b807237f302b824c5ad943b5e3adb68a79ebd5c29c51c30edc488b88deec82e8"
    request_sha = "249ab2cb0171bbf79c453ee9bf0310c4319866c41a65008a7d46d966c96fa255"

    def need(condition, message):
        if not condition:
            raise ValueError(message)

    def sha(raw):
        return hashlib.sha256(raw).hexdigest()

    def unique_object(pairs):
        result = {}
        for key, value in pairs:
            need(key not in result, "DUPLICATE_JSON_KEY: " + key)
            result[key] = value
        return result

    def invalid_constant(value):
        raise ValueError("INVALID_JSON_CONSTANT: " + value)

    def decode(raw):
        return json.loads(
            raw,
            object_pairs_hook=unique_object,
            parse_constant=invalid_constant
        )

    def read_receipt(receipt):
        path = (root / receipt["path"]).resolve()
        need(path.is_relative_to(root), "PATH_OUTSIDE_PROJECT")
        raw = path.read_bytes()
        need(sha(raw) == receipt["file_bytes_sha256"], "HASH_CONFLICT: " + str(path))
        return raw

    def local_receipt(stem, digest):
        path = directory / (stem + "." + digest + ".json")
        return {
            "path": str(path.relative_to(root)),
            "file_bytes_sha256": digest
        }

    freeze_receipt = local_receipt("C024.R10_selected_evidence_freeze", freeze_sha)
    request_receipt = local_receipt("C037.R11_response_regression_inputs", request_sha)
    freeze = decode(read_receipt(freeze_receipt))
    request = decode(read_receipt(request_receipt))
    need(
        freeze["selected_method"]["variant"] == request["selected_variant"]
        == "char_rrf",
        "SELECTED_VARIANT_CONFLICT"
    )

    tasks = request["selected_version_execution_requests"]
    task_ids = {task["regression_spec_id"] for task in tasks}
    need(len(tasks) == len(task_ids) == 36, "SCENARIO_POPULATION_CONFLICT")
    need(len({task["candidate_id"] for task in tasks}) == 34,
         "RESPONSE_POPULATION_CONFLICT")
    need((bundle_path is None) == (bundle_sha is None), "INCOMPLETE_BUNDLE_RECEIPT")

    bundle_receipt = None
    runtime_receipt = None
    assessment_index = {}

    if bundle_path is not None:
        bundle_receipt = {"path": bundle_path, "file_bytes_sha256": bundle_sha}
        bundle = decode(read_receipt(bundle_receipt))
        runtime_receipt = bundle["runtime_receipt"]
        runtime = decode(read_receipt(runtime_receipt))
        need(
            runtime["selected_freeze_sha256"] == freeze_sha
            and runtime["request_set_sha256"] == request_sha
            and runtime["variant"] == "char_rrf",
            "RUNTIME_PROVENANCE_CONFLICT"
        )
        for receipt in bundle["assessment_receipts"]:
            assessment = decode(read_receipt(receipt))
            spec_id = assessment["regression_spec_id"]
            need(spec_id in task_ids, "ASSESSMENT_OUTSIDE_REQUEST_SET: " + spec_id)
            need(spec_id not in assessment_index, "DUPLICATE_ASSESSMENT: " + spec_id)
            assessment_index[spec_id] = (assessment, receipt)

    outcomes = []
    response_receipts_by_candidate = {}

    for task in tasks:
        spec_id = task["regression_spec_id"]
        candidate_id = task["candidate_id"]
        outcome = {
            "regression_spec_id": spec_id,
            "candidate_id": candidate_id,
            "kind": task["kind"],
            "status": "BLOCKED",
            "check_executed": False
        }
        entry = assessment_index.get(spec_id)
        if entry is None:
            outcome["reason"] = "MATCHED_RUNTIME_RESPONSE_AND_ASSESSMENT_REQUIRED"
            outcome["required_execution_evidence"] = task["required_execution_evidence"]
            outcomes.append(outcome)
            continue

        assessment, assessment_receipt = entry
        response_receipt = assessment["response_receipt"]
        evidence_receipt = assessment["evidence_receipt"]
        response = decode(read_receipt(response_receipt))
        evidence = decode(read_receipt(evidence_receipt))

        need(
            assessment["candidate_id"] == response["candidate_id"]
            == evidence["candidate_id"] == candidate_id
            and assessment["runtime_receipt_sha256"]
            == runtime_receipt["file_bytes_sha256"]
            and response["runtime_receipt_sha256"]
            == runtime_receipt["file_bytes_sha256"]
            and evidence["runtime_receipt_sha256"]
            == runtime_receipt["file_bytes_sha256"]
            and response["evidence_bundle_sha256"]
            == evidence_receipt["file_bytes_sha256"],
            spec_id + ": RESPONSE_EVIDENCE_OR_RUNTIME_BINDING_CONFLICT"
        )
        response_sha = response_receipt["file_bytes_sha256"]
        if candidate_id in response_receipts_by_candidate:
            need(
                response_receipts_by_candidate[candidate_id] == response_sha,
                spec_id + ": MULTIPLE_RESPONSES_FOR_ONE_CANDIDATE"
            )
        response_receipts_by_candidate[candidate_id] = response_sha

        outcome["assessment_receipt"] = assessment_receipt
        outcome["response_receipt"] = response_receipt
        outcome["evidence_receipt"] = evidence_receipt
        outcome["reviewer_provenance"] = assessment["reviewer_provenance"]

        if assessment.get("scenario_scope_accepted") is not True:
            outcome["reason"] = "ASSESSMENT_SCOPE_REQUIRES_QUALIFICATION"
            outcomes.append(outcome)
            continue

        if task["kind"] == "ANSWER_DECISION":
            expected = task["behavior_expectation"]["expected_behavior"]
            need(
                assessment["expected_behavior"] == expected,
                spec_id + ": EXPECTED_BEHAVIOR_CONFLICT"
            )
            observed = assessment["observed_behavior"]
            limit = assessment["required_limit_check"]
            need(limit in {"SATISFIED", "VIOLATED", "NOT_APPLICABLE", "UNDETERMINED"},
                 spec_id + ": UNKNOWN_LIMIT_CHECK")
            if observed == "UNDETERMINED" or limit == "UNDETERMINED":
                outcome["reason"] = "BEHAVIOR_OR_LIMIT_UNDETERMINED"
                outcomes.append(outcome)
                continue
            passed = observed == expected and limit in {"SATISFIED", "NOT_APPLICABLE"}
            outcome["expected_behavior"] = expected
            outcome["observed_behavior"] = observed
            outcome["required_limit_check"] = limit
        else:
            historical = task["historical_link_expectation"]
            need(task["kind"] == "CITATION_WITNESS", spec_id + ": UNKNOWN_KIND")
            need(
                assessment["target_link_mapping"] == "EXPLICITLY_BOUND"
                and assessment["historical_claim_index"] == historical["claim_index"]
                and assessment["evidence_id"] == historical["evidence_id"],
                spec_id + ": TARGET_LINK_BINDING_CONFLICT"
            )
            claim_index = assessment["selected_claim_index"]
            claims = response["response_document"]["claims"]
            need(type(claim_index) is int and 0 <= claim_index < len(claims),
                 spec_id + ": SELECTED_CLAIM_INDEX_CONFLICT")
            evidence_id = assessment["evidence_id"]
            support = assessment["support"]
            need(support in {"SUPPORTED", "NOT_SUPPORTED", "UNDETERMINED"},
                 spec_id + ": UNKNOWN_LINK_SUPPORT")
            cited = evidence_id in claims[claim_index]["evidence_ids"]
            if support == "SUPPORTED":
                sources = [
                    item for item in evidence["sources"]
                    if item["evidence_id"] == evidence_id
                ]
                need(len(sources) == 1, spec_id + ": EVIDENCE_SOURCE_NOT_UNIQUE")
                text = sources[0]["text"]
                spans = assessment["exact_spans"]
                passed = cited and bool(spans) and all(
                    0 <= span["start"] < span["end"] <= len(text)
                    and text[span["start"]:span["end"]] == span["quote"]
                    for span in spans
                )
            else:
                # انتظار این تست، ثبت صریح رأی نامعین یا عدم پشتیبانی را می‌پذیرد.
                passed = True
            outcome["support_as_assessed"] = support
            outcome["selected_claim_index"] = claim_index
            outcome["evidence_id"] = evidence_id
            outcome["link_is_cited"] = cited

        outcome["status"] = "PASS" if passed else "FAIL"
        outcome["check_executed"] = True
        outcome["scope"] = "VERSION_MATCHED_RECEIPT_AND_RECORDED_ASSESSMENT_CHECK"
        outcomes.append(outcome)

    counts = {
        status: sum(row["status"] == status for row in outcomes)
        for status in ("PASS", "FAIL", "BLOCKED")
    }
    blocked_ids = {
        row["regression_spec_id"] for row in outcomes if row["status"] == "BLOCKED"
    }

    result = {
        "schema_version": "NB22E_V3_C038_R11_V1",
        "status": "R11_RESPONSE_REGRESSION_RECEIPTS_CHECKED",
        "roadmap_action": "R11",
        "selected_variant": "char_rrf",
        "freeze_receipt": freeze_receipt,
        "request_receipt": request_receipt,
        "bundle_receipt": bundle_receipt,
        "runtime_receipt": runtime_receipt,
        "scenario_counts": counts,
        "distinct_selected_responses_bound": len(response_receipts_by_candidate),
        "scenario_checks_executed": counts["PASS"] + counts["FAIL"],
        "scenario_outcomes": outcomes,
        "outstanding_requests": [
            task for task in tasks if task["regression_spec_id"] in blocked_ids
        ],
        "fixed_pool_retrieval_scenarios_blocked": 260,
        "selected_version_gate": "BLOCKED",
        "all_R11_items_complete": False,
        "new_api_calls": 0
    }

    raw = (json.dumps(result, ensure_ascii=False, sort_keys=True,
                      indent=2, allow_nan=False) + "\n").encode("utf-8")
    digest = sha(raw)
    output = directory / ("C038.R11_response_regression_results." + digest + ".json")
    try:
        with output.open("xb") as handle:
            handle.write(raw)
    except FileExistsError:
        need(output.read_bytes() == raw, "OUTPUT_CONTENT_CONFLICT")

    saved = output.read_bytes()
    need(sha(saved) == digest and decode(saved) == result, "OUTPUT_RELOAD_CONFLICT")
    print(json.dumps({
        "status": result["status"],
        "scenario_counts": counts,
        "distinct_selected_responses_bound": len(response_receipts_by_candidate),
        "scenario_checks_executed": result["scenario_checks_executed"],
        "selected_version_gate": result["selected_version_gate"],
        "output": {
            "path": str(output.relative_to(root)),
            "file_bytes_sha256": digest,
            "reload_validation": "PASS"
        }
    }, ensure_ascii=False, indent=2))
    return result


V3_C038 = run_v3_c038(C038_RUN_BUNDLE_PATH, C038_RUN_BUNDLE_SHA256)
````

سپس ارزیابی رفتار پاسخ و پیوندهای استنادی است.

````python
# C039 — R11: ساخت ورودی واقعی تولید پاسخ برای ۳۴ پرونده
# imports از سل اول: Path, os, json, hashlib, math, re,
# unicodedata, collections, ast

def run_v3_c039():
    required = (
        "Path", "os", "json", "hashlib", "math", "re",
        "unicodedata", "collections", "ast"
    )
    missing = [name for name in required if name not in globals()]
    if missing:
        raise RuntimeError("ابتدا سل imports را اجرا کنید: " + ", ".join(missing))

    root = Path(os.environ.get(
        "RAG_FINANCE_ROOT",
        "/Users/macbookpro/AMIR_DATA/00_WORKS/01_PROJECTS/"
        "RAG-projects/RAG_finance"
    )).expanduser().resolve()
    directory = root / "reports/phase2/quality_improvement/NB22E.lab.v3"
    receipts = {}

    def need(condition, message):
        if not condition:
            raise ValueError(message)

    def sha(raw):
        return hashlib.sha256(raw).hexdigest()

    def canonical(value):
        return json.dumps(
            value, ensure_ascii=False, sort_keys=True,
            separators=(",", ":"), allow_nan=False
        ).encode("utf-8")

    def read_bound(path, digest, jsonl=False):
        resolved = (root / path).resolve()
        need(resolved.is_relative_to(root), "PATH_OUTSIDE_PROJECT: " + path)
        raw = resolved.read_bytes()
        need(sha(raw) == digest, "HASH_CONFLICT: " + path)
        receipts[path] = digest
        if jsonl:
            return [json.loads(line) for line in raw.splitlines() if line.strip()]
        return json.loads(raw)

    def read_v3(stem, digest):
        path = directory / (stem + "." + digest + ".json")
        return read_bound(str(path.relative_to(root)), digest)

    freeze_sha = "b807237f302b824c5ad943b5e3adb68a79ebd5c29c51c30edc488b88deec82e8"
    request_sha = "249ab2cb0171bbf79c453ee9bf0310c4319866c41a65008a7d46d966c96fa255"

    frozen = read_v3("C024.R10_selected_evidence_freeze", freeze_sha)
    requests = read_v3("C037.R11_response_regression_inputs", request_sha)
    previous = read_v3(
        "C038.R11_response_regression_results",
        "0c512dcc6d82560f0b6f5b4c6027938d2986e822d0268008dfdfd6beb6cd16c3"
    )
    diagnostic = read_v3(
        "C036.R11_expanded_pool_diagnostic",
        "69a6ffdbd965ce8a2a6176ba201445798d1604c84e73954a6a407b7411f2b7b2"
    )
    need(
        previous["request_receipt"]["file_bytes_sha256"] == request_sha
        and previous["freeze_receipt"]["file_bytes_sha256"] == freeze_sha
        and previous["scenario_counts"] == {"PASS": 0, "FAIL": 0, "BLOCKED": 36},
        "C038_INPUT_OR_STATUS_CONFLICT"
    )

    selected = frozen["selected_method"]
    need(selected["variant"] == requests["selected_variant"] == "char_rrf",
         "SELECTED_VARIANT_CONFLICT")
    configuration = selected["configuration"]
    code_cells = selected["producer_code_snapshot"]
    need(
        sha(canonical(code_cells)) == selected["producer_code_sha256"]
        and sha(canonical(configuration))
        == selected["configuration_canonical_sha256"],
        "FROZEN_PRODUCER_CONFLICT"
    )
    variants = [
        row for row in configuration["ranking_specs"]
        if row["name"] == "char_rrf"
    ]
    need(len(variants) == 1 and variants[0]["channels"] == ["char"],
         "CHAR_RRF_CONFIGURATION_CONFLICT")

    tasks_by_candidate = {}
    task_ids = set()
    for task in requests["selected_version_execution_requests"]:
        spec_id = task["regression_spec_id"]
        need(spec_id not in task_ids, "DUPLICATE_SCENARIO: " + spec_id)
        task_ids.add(spec_id)
        tasks_by_candidate.setdefault(task["candidate_id"], []).append(task)
    need(len(task_ids) == 36 and len(tasks_by_candidate) == 34,
         "REQUEST_POPULATION_CONFLICT")
    wanted = set(tasks_by_candidate)

    def unique_index(rows, key):
        indexed = {}
        for row in rows:
            identity = row[key]
            need(identity not in indexed, "DUPLICATE_" + key + ": " + identity)
            indexed[identity] = row
        return indexed

    bindings = frozen["evaluation_lock"]["data_bindings"]
    chunks = unique_index(read_bound(
        bindings["chunks"]["path"], bindings["chunks"]["file_bytes_sha256"],
        jsonl=True
    ), "chunk_id")
    questions = unique_index([
        row for row in read_bound(
            bindings["golden"]["path"], bindings["golden"]["file_bytes_sha256"],
            jsonl=True
        )
        if row["candidate_id"] in wanted
    ], "candidate_id")

    rank_path = "data/phase2/retrieval/reranker_v1/reranked_candidates_v1.jsonl"
    historical = unique_index([
        row for row in read_bound(
            rank_path, diagnostic["input_receipts"][rank_path], jsonl=True
        )
        if row["candidate_id"] in wanted
    ], "candidate_id")
    need(set(questions) == set(historical) == wanted,
         "GOLDEN_OR_HISTORICAL_RANK_POPULATION_CONFLICT")

    # فقط تعریف پنج تابع قبلاً بررسی‌شده از Snapshot فریزشده بار می‌شود.
    names = {
        "normalized_words", "char_tokens", "build_index",
        "lexical_search", "lexical_fusion"
    }
    functions = {}
    for cell in code_cells:
        for node in ast.parse(cell).body:
            if isinstance(node, ast.FunctionDef) and node.name in names:
                need(
                    node.name not in functions and not node.decorator_list
                    and not node.args.defaults and not any(node.args.kw_defaults),
                    "UNEXPECTED_FROZEN_FUNCTION_DEFINITION: " + node.name
                )
                functions[node.name] = node
    need(set(functions) == names, "FROZEN_RETRIEVAL_FUNCTIONS_MISSING")

    namespace = {
        "CONFIG": configuration,
        "collections": collections,
        "math": math,
        "re": re,
        "unicodedata": unicodedata,
        "require": need
    }
    module = ast.Module(body=list(functions.values()), type_ignores=[])
    exec(compile(module, "<pinned-NB22D-retrieval-functions>", "exec"), namespace)
    index = namespace["build_index"](chunks, "char")

    prepared = []
    for candidate_id in sorted(wanted):
        historical_row = historical[candidate_id]
        candidates = historical_row["candidates"]
        need(
            historical_row["split"] == "test"
            and [row["rank"] for row in candidates] == list(range(1, 51)),
            candidate_id + ": HISTORICAL_POOL50_CONFLICT"
        )
        baseline = [row["chunk_id"] for row in candidates]
        need(len(set(baseline)) == 50 and set(baseline).issubset(chunks),
             candidate_id + ": BASELINE_CHUNK_ID_CONFLICT")

        question = questions[candidate_id]["evaluation_content"]["question"]
        need(isinstance(question, str) and bool(question.strip()),
             candidate_id + ": EMPTY_QUESTION")
        lexical = namespace["lexical_search"](index, question, "char")
        order = namespace["lexical_fusion"](
            baseline, {"char": lexical}, variants[0]
        )
        need(
            len(order) == len(set(order))
            and set(order) == set(baseline) | set(lexical)
            and len(order) >= 10,
            candidate_id + ": FUSION_MEMBERSHIP_CONFLICT"
        )

        evidence = [{
            "rank": rank,
            "evidence_id": chunk_id,
            "text": chunks[chunk_id]["content"],
            "text_sha256": sha(chunks[chunk_id]["content"].encode("utf-8"))
        } for rank, chunk_id in enumerate(order[:10], start=1)]

        generation_input = {
            "candidate_id": candidate_id,
            "question": question,
            "sources": evidence
        }
        prepared.append({
            "candidate_id": candidate_id,
            "generation_input": generation_input,
            "generation_input_sha256": sha(canonical(generation_input)),
            "scenario_expectations": tasks_by_candidate[candidate_id],
            "historical_baseline_top10": baseline[:10],
            "expanded_candidate_count": len(order)
        })

    need(len(prepared) == 34, "PREPARED_RESPONSE_COUNT_CONFLICT")
    result = {
        "schema_version": "NB22E_V3_C039_R11_V1",
        "status": "R11_DIAGNOSTIC_GENERATION_INPUTS_PREPARED",
        "roadmap_action": "R11",
        "scope": "FROZEN_CHAR_FUSION_ON_HISTORICAL_POOL50_INCUMBENT",
        "source_freeze_sha256": freeze_sha,
        "source_scenario_request_sha256": request_sha,
        "input_receipts": receipts,
        "prepared_generation_input_count": len(prepared),
        "scenario_count": len(task_ids),
        "generation_inputs": prepared,
        "responses_generated": 0,
        "product_regression_tests_executed": 0,
        "new_api_calls": 0,
        "selected_version_gate": "BLOCKED",
        "next_step": "BIND_GENERATOR_AND_RECORD_RESPONSES_FOR_THE_PREPARED_INPUTS"
    }

    payload = (json.dumps(
        result, ensure_ascii=False, sort_keys=True, indent=2, allow_nan=False
    ) + "\n").encode("utf-8")
    digest = sha(payload)
    output = directory / ("C039.R11_diagnostic_generation_inputs." + digest + ".json")
    try:
        with output.open("xb") as handle:
            handle.write(payload)
    except FileExistsError:
        need(output.read_bytes() == payload, "OUTPUT_CONTENT_CONFLICT")

    saved = output.read_bytes()
    need(sha(saved) == digest and json.loads(saved) == result,
         "OUTPUT_RELOAD_CONFLICT")
    print(json.dumps({
        "status": result["status"],
        "prepared_generation_inputs": len(prepared),
        "scenario_count": len(task_ids),
        "responses_generated": 0,
        "selected_version_gate": "BLOCKED",
        "output": {
            "path": str(output.relative_to(root)),
            "file_bytes_sha256": digest,
            "reload_validation": "PASS"
        }
    }, ensure_ascii=False, indent=2))
    return result


V3_C039 = run_v3_c039()
````

### کار یازده — ثبت طرح و تولید پاسخ تشخیصی

ورودی این سل، ۳۴ بستهٔ سؤال و شاهد ثبت‌شده در C039 است.

قالب پیام مولد آروان به کد پین‌شده متصل می‌شود.
هر درخواست، مدل، هش ورودی، ترتیب شاهدها و سقف خروجی مشخص دارد.

مقدار صفر برای حد درخواست، فقط طرح و رسیدها را آماده می‌کند.
پس از تأیید هزینه، پایلوت یک‌پرونده‌ای اجرا می‌شود.

Attempt پیش از ارسال ثبت می‌شود.
پاسخ خام، متن پاسخ، مصرف گزارش‌شده و مدت فراخوانی ذخیره می‌شوند.
Attempt فاقد نتیجه در وضعیت نامعلوم می‌ماند.

دامنهٔ پاسخ‌ها، اجرای تشخیصی روی زمینهٔ C039 است.
گام بعدی، ارزیابی پاسخ‌های ثبت‌شده بر اساس انتظار سناریوهاست.

````python
# C040 — R11: طرح و اجرای محدود تولید پاسخ روی ورودی C039
# imports از سل اول: Path, os, json, hashlib, ast, importlib, time

C040_NEW_REQUEST_LIMIT = 0  # پس از تأیید هزینهٔ پایلوت، 1 بگذارید.


def run_v3_c040(new_request_limit):
    required = ("Path", "os", "json", "hashlib", "ast", "importlib", "time")
    missing = [name for name in required if name not in globals()]
    if missing:
        raise RuntimeError("ایمپورت لازم در سل اول: " + ", ".join(missing))

    def need(condition, message):
        if not condition:
            raise ValueError(message)

    def canonical(value):
        return json.dumps(
            value, ensure_ascii=False, sort_keys=True,
            separators=(",", ":"), allow_nan=False
        ).encode("utf-8")

    def sha(raw):
        return hashlib.sha256(raw).hexdigest()

    def write_once(path, value):
        payload = canonical(value) + b"\n"
        try:
            with path.open("xb") as handle:
                handle.write(payload)
        except FileExistsError:
            need(path.read_bytes() == payload, "EXISTING_FILE_CONFLICT: " + str(path))
        saved = path.read_bytes()
        need(saved == payload and json.loads(saved) == value,
             "RELOAD_CONFLICT: " + str(path))
        return {
            "path": str(path.relative_to(root)),
            "file_bytes_sha256": sha(saved)
        }

    need(type(new_request_limit) is int and 0 <= new_request_limit <= 34,
         "REQUEST_LIMIT_OUT_OF_RANGE")

    root = Path(os.environ.get(
        "RAG_FINANCE_ROOT",
        "/Users/macbookpro/AMIR_DATA/00_WORKS/01_PROJECTS/"
        "RAG-projects/RAG_finance"
    )).expanduser().resolve()
    directory = root / "reports/phase2/quality_improvement/NB22E.lab.v3"
    input_sha = "2b07138f01ba855cc9fb00df6b79df2efa6e98906c529cc3955ccff46fc68a38"
    input_path = directory / (
        "C039.R11_diagnostic_generation_inputs." + input_sha + ".json"
    )
    input_raw = input_path.read_bytes()
    need(sha(input_raw) == input_sha, "C039_HASH_CONFLICT")
    source = json.loads(input_raw)
    need(
        source["scope"] == "FROZEN_CHAR_FUSION_ON_HISTORICAL_POOL50_INCUMBENT"
        and source["prepared_generation_input_count"] == 34
        and source["scenario_count"] == 36,
        "C039_SCOPE_OR_POPULATION_CONFLICT"
    )

    # خواندن قالب پیام از کد پین‌شدهٔ مولد، بدون import یا اجرای فایل.
    producer_path = root / "src/generation/arvan_generator.py"
    producer_raw = producer_path.read_bytes()
    producer_blob = hashlib.sha1(
        b"blob " + str(len(producer_raw)).encode("ascii") + b"\0" + producer_raw
    ).hexdigest()
    need(producer_blob == "db142ed6e1bf65a51e1548feb99e02a58cce0c80",
         "ARVAN_GENERATOR_SOURCE_DIFFERS_FROM_REVIEWED_BLOB")

    tree = ast.parse(producer_raw)
    producer_class = next(
        node for node in tree.body
        if isinstance(node, ast.ClassDef) and node.name == "ArvanGenerator"
    )
    method = next(
        node for node in producer_class.body
        if isinstance(node, ast.FunctionDef) and node.name == "generate_answer"
    )
    prompt_node = next(
        node.value for node in method.body
        if isinstance(node, ast.Assign) and any(
            isinstance(target, ast.Name) and target.id == "system_prompt"
            for target in node.targets
        )
    )
    system_prompt = ast.literal_eval(prompt_node).strip()

    model = os.environ.get("ARVAN_CHAT_MODEL") or "Qwen3-30B-A3B"
    endpoint = os.environ.get("ARVAN_CHAT_URL")
    settings = {
        "model": model,
        "temperature": 0.2,
        "max_tokens": 600,
        "max_retries": 0,
        "timeout_seconds": 120,
        "endpoint_sha256": sha(endpoint.encode("utf-8")) if endpoint else None
    }
    protocol = "C040_ARVAN_DIAGNOSTIC_C039_V1"
    jobs = []
    candidate_ids = set()
    scenario_ids = set()

    for entry in source["generation_inputs"]:
        data = entry["generation_input"]
        candidate_id = entry["candidate_id"]
        need(
            candidate_id == data["candidate_id"]
            and candidate_id not in candidate_ids
            and sha(canonical(data)) == entry["generation_input_sha256"],
            "C039_GENERATION_INPUT_CONFLICT: " + candidate_id
        )
        candidate_ids.add(candidate_id)
        sources = data["sources"]
        need(
            [item["rank"] for item in sources] == list(range(1, 11))
            and len({item["evidence_id"] for item in sources}) == 10,
            "C039_SOURCE_ORDER_CONFLICT: " + candidate_id
        )
        for item in sources:
            need(sha(item["text"].encode("utf-8")) == item["text_sha256"],
                 "C039_EVIDENCE_TEXT_HASH_CONFLICT: " + candidate_id)

        context = ("\n\n" + "-" * 80 + "\n\n").join(
            f'[منبع {item["rank"]}]\n'
            f'شناسه چانک: {item["evidence_id"]}\n\nمتن:\n{item["text"]}'
            for item in sources
        )
        user_prompt = f'سؤال:\n{data["question"]}\n\nمنابع:\n{context}\n\nپاسخ:'
        request = {
            "model": model,
            "temperature": settings["temperature"],
            "max_tokens": settings["max_tokens"],
            "messages": [
                {"role": "system", "content": system_prompt},
                {"role": "user", "content": user_prompt}
            ]
        }
        spec_ids = []
        for task in entry["scenario_expectations"]:
            spec_id = task["regression_spec_id"]
            need(task["candidate_id"] == candidate_id and spec_id not in scenario_ids,
                 "SCENARIO_IDENTITY_CONFLICT: " + spec_id)
            scenario_ids.add(spec_id)
            spec_ids.append(spec_id)

        job = {
            "protocol": protocol,
            "input_file_sha256": input_sha,
            "candidate_id": candidate_id,
            "generation_input_sha256": entry["generation_input_sha256"],
            "regression_spec_ids": sorted(spec_ids),
            "ordered_evidence_ids": [item["evidence_id"] for item in sources],
            "generator_blob_sha1": producer_blob,
            "settings": settings,
            "request": request
        }
        jobs.append({"job_sha256": sha(canonical(job)), **job})

    need(len(candidate_ids) == len(jobs) == 34 and len(scenario_ids) == 36,
         "JOB_POPULATION_CONFLICT")

    plan = {
        "schema_version": "NB22E_V3_C040_PLAN_V1",
        "scope": source["scope"],
        "protocol": protocol,
        "input_receipt": {
            "path": str(input_path.relative_to(root)),
            "file_bytes_sha256": input_sha
        },
        "settings": settings,
        "planned_requests": len(jobs),
        "maximum_requested_output_tokens": len(jobs) * settings["max_tokens"],
        "jobs": jobs
    }
    plan_sha = sha(canonical(plan))
    plan_receipt = write_once(
        directory / ("C040.R11_generation_plan." + plan_sha + ".json"), plan
    )

    client = None
    if new_request_limit:
        api_key = os.environ.get("ARVAN_CHAT_API_KEY") or os.environ.get("ARVAN_AI_API_KEY")
        need(bool(endpoint and endpoint.startswith("https://") and api_key),
             "ARVAN_HTTPS_ENDPOINT_AND_API_KEY_REQUIRED")
        client = importlib.import_module("openai").OpenAI(
            api_key=api_key,
            base_url=endpoint,
            max_retries=settings["max_retries"],
            timeout=settings["timeout_seconds"]
        )

    job_root = directory / "C040.R11_generation_jobs"
    job_root.mkdir(exist_ok=True)
    outcomes = []
    attempted = 0
    stop_dispatch = False

    for job in jobs:
        job_sha = job["job_sha256"]
        attempt_path = job_root / (job_sha + ".attempt.json")
        response_path = job_root / (job_sha + ".response.json")
        outcome = {
            "candidate_id": job["candidate_id"],
            "job_sha256": job_sha,
            "regression_spec_ids": job["regression_spec_ids"]
        }

        if response_path.is_file():
            raw = response_path.read_bytes()
            saved_response = json.loads(raw)
            need(
                saved_response["job_sha256"] == job_sha
                and saved_response["generation_input_sha256"]
                == job["generation_input_sha256"],
                "SAVED_RESPONSE_BINDING_CONFLICT: " + job_sha
            )
            outcome.update({
                "status": "RESPONSE_SAVED",
                "response_receipt": {
                    "path": str(response_path.relative_to(root)),
                    "file_bytes_sha256": sha(raw)
                }
            })
        elif attempt_path.exists():
            outcome["status"] = "ATTEMPT_OUTCOME_UNKNOWN"
        elif client is None or attempted >= new_request_limit or stop_dispatch:
            outcome["status"] = "READY_FOR_AUTHORIZED_GENERATION"
        else:
            attempt = {
                "job_sha256": job_sha,
                "plan_receipt": plan_receipt,
                "candidate_id": job["candidate_id"],
                "request_sha256": sha(canonical(job["request"]))
            }
            try:
                # قفل ایجاد انحصاری جلوی ارسال هم‌زمان همان Job را می‌گیرد.
                with attempt_path.open("xb") as handle:
                    handle.write(canonical(attempt) + b"\n")
                    handle.flush()
                    os.fsync(handle.fileno())
            except FileExistsError:
                outcome["status"] = "ATTEMPT_OUTCOME_UNKNOWN"
                outcomes.append(outcome)
                continue

            attempted += 1
            started = time.monotonic()
            try:
                completion = client.chat.completions.create(**job["request"])
                response_receipt = write_once(response_path, {
                    "job_sha256": job_sha,
                    "candidate_id": job["candidate_id"],
                    "generation_input_sha256": job["generation_input_sha256"],
                    "answer_text": completion.choices[0].message.content,
                    "finish_reason": completion.choices[0].finish_reason,
                    "elapsed_seconds": time.monotonic() - started,
                    "raw_provider_response": completion.model_dump(mode="json")
                })
            except Exception as error:
                outcome["status"] = "ATTEMPT_OUTCOME_UNKNOWN"
                outcome["error_type"] = type(error).__name__
                stop_dispatch = True
            else:
                outcome["status"] = "RESPONSE_SAVED"
                outcome["response_receipt"] = response_receipt

        outcomes.append(outcome)

    counts = {}
    for row in outcomes:
        counts[row["status"]] = counts.get(row["status"], 0) + 1
    need(sum(counts.values()) == 34, "JOB_ACCOUNTING_CONFLICT")

    result = {
        "schema_version": "NB22E_V3_C040_R11_V1",
        "roadmap_action": "R11",
        "scope": source["scope"],
        "plan_receipt": plan_receipt,
        "status_counts": counts,
        "new_api_calls_attempted": attempted,
        "response_count": counts.get("RESPONSE_SAVED", 0),
        "product_regression_tests_executed": 0,
        "selected_version_gate": "BLOCKED",
        "outcomes": outcomes
    }
    result_sha = sha(canonical(result))
    output = write_once(
        directory / ("C040.R11_generation_execution." + result_sha + ".json"), result
    )
    print(json.dumps({
        "status_counts": counts,
        "new_api_calls_attempted": attempted,
        "response_count": result["response_count"],
        "selected_version_gate": "BLOCKED",
        "plan_receipt": plan_receipt,
        "output": {**output, "reload_validation": "PASS"}
    }, ensure_ascii=False, indent=2))
    return result


V3_C040 = run_v3_c040(C040_NEW_REQUEST_LIMIT)
````

### کار یازده — پایلوت یک پاسخ تشخیصی

اولین درخواست از طرح هش‌شدهٔ C040 انتخاب می‌شود.
ورودی، مدل، مقصد و سقف ۶۰۰ توکن خروجی با طرح تطبیق می‌یابند.

پس از تأیید هزینه، یک درخواست ارسال می‌شود.
رسید Attempt پیش از ارسال ثبت می‌شود و پاسخ خام،
متن پاسخ، دلیل پایان و مصرف گزارش‌شده ذخیره می‌شوند.

اگر پاسخ قبلاً ثبت شده باشد، همان رسید خوانده می‌شود.
اگر Attempt قبلی نتیجهٔ ثبت‌شده نداشته باشد،
ابتدا باید وضعیت همان Attempt روشن شود.

گام بعدی، ارزیابی پاسخ پایلوت با انتظار سناریوی مرتبط است.

</div>

````python
# C041 — پایلوت دقیقاً یک Job از طرح ثبت‌شدهٔ C040
# imports از سل اول: Path, os, json, hashlib, importlib, time

C041_APPROVE_ONE_REQUEST = False


def run_v3_c041(approved):
    root = Path(os.environ.get(
        "RAG_FINANCE_ROOT",
        "/Users/macbookpro/AMIR_DATA/00_WORKS/01_PROJECTS/"
        "RAG-projects/RAG_finance"
    )).expanduser().resolve()
    directory = root / "reports/phase2/quality_improvement/NB22E.lab.v3"

    def need(ok, message):
        if not ok:
            raise ValueError(message)

    def sha(raw):
        return hashlib.sha256(raw).hexdigest()

    def canonical(value):
        return json.dumps(
            value, ensure_ascii=False, sort_keys=True,
            separators=(",", ":"), allow_nan=False
        ).encode("utf-8")

    def receipt(path):
        return {
            "path": str(path.relative_to(root)),
            "file_bytes_sha256": sha(path.read_bytes())
        }

    plan_path = directory / (
        "C040.R11_generation_plan."
        "608d2551794d2028f973e23a89e96079b0f8b2fead857f43c16079379f29b3bd.json"
    )
    plan_raw = plan_path.read_bytes()
    need(
        sha(plan_raw)
        == "cda83957141dcfbf85e2073b9af21e65d8406da2e328da6c2b5f25a713579e85",
        "C040_PLAN_FILE_HASH_CONFLICT"
    )
    plan = json.loads(plan_raw)
    need(
        plan["scope"] == "FROZEN_CHAR_FUSION_ON_HISTORICAL_POOL50_INCUMBENT"
        and plan["planned_requests"] == len(plan["jobs"]) == 34,
        "C040_PLAN_SCOPE_OR_POPULATION_CONFLICT"
    )

    job = plan["jobs"][0]
    job_sha = job["job_sha256"]
    need(
        sha(canonical({
            key: value for key, value in job.items() if key != "job_sha256"
        })) == job_sha,
        "PILOT_JOB_HASH_CONFLICT"
    )

    request = job["request"]
    settings = job["settings"]
    need(
        settings == plan["settings"]
        and request["model"] == settings["model"]
        and request["max_tokens"] == settings["max_tokens"] == 600
        and settings["max_retries"] == 0,
        "PILOT_REQUEST_SETTINGS_CONFLICT"
    )

    jobs_dir = directory / "C040.R11_generation_jobs"
    jobs_dir.mkdir(exist_ok=True)
    attempt_path = jobs_dir / (job_sha + ".attempt.json")
    response_path = jobs_dir / (job_sha + ".response.json")
    attempted = 0
    response_receipt = None
    finish_reason = None
    usage = None
    error_type = None

    if response_path.is_file():
        status = "EXISTING_PILOT_RESPONSE_REUSED"
    elif attempt_path.exists():
        status = "PRIOR_PILOT_ATTEMPT_OUTCOME_UNKNOWN"
    elif approved is not True:
        status = "PILOT_READY_AWAITING_COST_APPROVAL"
    else:
        endpoint = os.environ.get("ARVAN_CHAT_URL")
        api_key = (
            os.environ.get("ARVAN_CHAT_API_KEY")
            or os.environ.get("ARVAN_AI_API_KEY")
        )
        need(
            bool(endpoint and endpoint.startswith("https://") and api_key),
            "ARVAN_ENDPOINT_OR_API_KEY_MISSING"
        )
        need(
            sha(endpoint.encode("utf-8")) == settings["endpoint_sha256"],
            "ENDPOINT_DIFFERS_FROM_C040_PLAN"
        )
        client = importlib.import_module("openai").OpenAI(
            api_key=api_key,
            base_url=endpoint,
            max_retries=0,
            timeout=settings["timeout_seconds"]
        )
        attempt = {
            "job_sha256": job_sha,
            "plan_receipt": receipt(plan_path),
            "candidate_id": job["candidate_id"],
            "request_sha256": sha(canonical(request))
        }

        try:
            with attempt_path.open("xb") as handle:
                handle.write(canonical(attempt) + b"\n")
                handle.flush()
                os.fsync(handle.fileno())
        except FileExistsError:
            status = "PRIOR_PILOT_ATTEMPT_OUTCOME_UNKNOWN"
        else:
            attempted = 1
            started = time.monotonic()
            try:
                completion = client.chat.completions.create(**request)
                provider_response = completion.model_dump(mode="json")
                choice = completion.choices[0] if completion.choices else None
                response = {
                    "job_sha256": job_sha,
                    "candidate_id": job["candidate_id"],
                    "generation_input_sha256": job["generation_input_sha256"],
                    "answer_text": choice.message.content if choice else None,
                    "finish_reason": choice.finish_reason if choice else None,
                    "elapsed_seconds": time.monotonic() - started,
                    "raw_provider_response": provider_response
                }
                payload = canonical(response) + b"\n"
                with response_path.open("xb") as handle:
                    handle.write(payload)
                    handle.flush()
                    os.fsync(handle.fileno())
                need(response_path.read_bytes() == payload, "RESPONSE_RELOAD_CONFLICT")
                status = "PILOT_RESPONSE_SAVED"
            except Exception as error:
                error_type = type(error).__name__
                status = "PILOT_ATTEMPT_REQUIRES_RECONCILIATION"
        finally:
            client.close()

    usable_answer = False
    if status in {"PILOT_RESPONSE_SAVED", "EXISTING_PILOT_RESPONSE_REUSED"}:
        response = json.loads(response_path.read_bytes())
        need(
            response["job_sha256"] == job_sha
            and response["candidate_id"] == job["candidate_id"]
            and response["generation_input_sha256"] == job["generation_input_sha256"],
            "PILOT_RESPONSE_BINDING_CONFLICT"
        )
        response_receipt = receipt(response_path)
        finish_reason = response["finish_reason"]
        usage = response["raw_provider_response"].get("usage")
        answer = response["answer_text"]
        usable_answer = (
            finish_reason == "stop"
            and isinstance(answer, str)
            and bool(answer.strip())
        )

    result = {
        "status": status,
        "candidate_id": job["candidate_id"],
        "regression_spec_ids": job["regression_spec_ids"],
        "job_sha256": job_sha,
        "requested_model": request["model"],
        "maximum_output_tokens": request["max_tokens"],
        "new_api_calls_attempted": attempted,
        "response_receipt": response_receipt,
        "finish_reason": finish_reason,
        "usage_as_reported": usage,
        "usable_answer_for_next_review": usable_answer,
        "error_type": error_type,
        "scope": plan["scope"],
        "selected_version_gate": "BLOCKED"
    }
    print(json.dumps(result, ensure_ascii=False, indent=2))
    return result


V3_C041 = run_v3_c041(C041_APPROVE_ONE_REQUEST)
````

````python
# Run after the notebook's first imports cell: Path, os, json, hashlib.

def prepare_r11_deepseek_pilot():
    root = Path(os.environ.get(
        "RAG_FINANCE_ROOT",
        "/Users/macbookpro/AMIR_DATA/00_WORKS/01_PROJECTS/"
        "RAG-projects/RAG_finance",
    )).expanduser().resolve()

    plan_path = Path(
        "reports/phase2/quality_improvement/NB22E.lab.v3/"
        "C040.R11_generation_plan."
        "608d2551794d2028f973e23a89e96079b0f8b2fead857f43c16079379f29b3bd.json"
    )
    expected_file_sha = (
        "cda83957141dcfbf85e2073b9af21e65d8406da2e328da6c2b5f25a713579e85"
    )
    candidate_id = "golden_v2_candidate_1c458d1b58d351f6ab1b32a6"
    scenario_id = "REG_c05a077fb1b2f216431f2cde"
    previous_job_sha = (
        "2b4d9896d7de02885046631d3bcf8bc749bc93a54ac2c421d98ee7d7e327ae11"
    )

    raw_plan = (root / plan_path).read_bytes()
    if hashlib.sha256(raw_plan).hexdigest() != expected_file_sha:
        raise ValueError("C040_PLAN_FILE_HASH_MISMATCH")

    plan = json.loads(raw_plan)
    matches = [
        job for job in plan["jobs"]
        if job.get("candidate_id") == candidate_id
        and job.get("job_sha256") == previous_job_sha
    ]
    if len(matches) != 1:
        raise ValueError("EXPECTED_EXACTLY_ONE_PINNED_C040_PILOT_JOB")

    original_job = matches[0]
    messages = original_job["request"]["messages"]
    if not isinstance(messages, list) or not messages:
        raise ValueError("C040_REQUEST_MESSAGES_MISSING")

    request = {
        "model": "DeepSeek-V4-Flash",
        "messages": messages,
        "temperature": 0,
        "max_tokens": 600,
        "thinking": {"type": "disabled"},
    }
    pilot = {
        "profile": "R11_DEEPSEEK_SINGLE_DIAGNOSTIC_PILOT_V1",
        "scope": "FROZEN_CHAR_FUSION_ON_HISTORICAL_POOL50_INCUMBENT",
        "candidate_id": candidate_id,
        "regression_spec_ids": [scenario_id],
        "generation_input_sha256": original_job["generation_input_sha256"],
        "source_plan_file_bytes_sha256": expected_file_sha,
        "source_qwen_job_sha256": previous_job_sha,
        "request": request,
    }
    canonical = json.dumps(
        pilot, ensure_ascii=False, sort_keys=True, separators=(",", ":"),
        allow_nan=False,
    ).encode("utf-8")
    pilot["pilot_sha256"] = hashlib.sha256(canonical).hexdigest()

    print(json.dumps({
        "status": "DEEPSEEK_PILOT_PREPARED_OFFLINE",
        "candidate_id": candidate_id,
        "regression_spec_ids": pilot["regression_spec_ids"],
        "pilot_sha256": pilot["pilot_sha256"],
        "requested_model": request["model"],
        "maximum_output_tokens": request["max_tokens"],
        "message_count": len(messages),
        "new_api_calls_attempted": 0,
        "response_count": 0,
        "selected_version_gate": "BLOCKED",
    }, ensure_ascii=False, indent=2))
    return pilot


V3_R11_DEEPSEEK_PILOT = prepare_r11_deepseek_pilot()
````

### C042 — اجرای تجمیعی DeepSeek برای کار ۱۱

این مرحله پاسخ و رسید مصرف را ثبت می‌کند. داوری سناریوها در کار ۱۱
باقی می‌ماند و وضعیت Gate نسخهٔ منتخب BLOCKED است.

````python
# C042 | R11 | Replace the entire previous C042 code cell.
# C001 imports: Path, os, json, hashlib, urllib.parse, urllib.request.
# Add `from dotenv import dotenv_values` to C001 if it is absent.
# C041 must have created V3_R11_DEEPSEEK_PILOT.

# Historical project rates, in toman per million tokens.
C042_INPUT_RATE = 43_680
C042_OUTPUT_RATE = 87_360
C042_BUDGET_TOMAN = 100_000


def run_v3_c042():
    def require(condition, code):
        if not condition:
            raise ValueError(code)

    def encode(value):
        return json.dumps(
            value, ensure_ascii=False, sort_keys=True,
            separators=(",", ":"), allow_nan=False,
        ).encode("utf-8")

    def sha(data):
        return hashlib.sha256(data).hexdigest()

    def write_once(path, record):
        with path.open("xb") as handle:
            handle.write(encode(record))
            handle.flush()
            os.fsync(handle.fileno())

    input_rate = C042_INPUT_RATE
    output_rate = C042_OUTPUT_RATE
    budget = C042_BUDGET_TOMAN * 1_000_000
    require(
        all(type(value) is int and value > 0
            for value in (input_rate, output_rate, C042_BUDGET_TOMAN)),
        "C042_INVALID_BUDGET_CONFIGURATION",
    )

    root = Path(os.environ.get(
        "RAG_FINANCE_ROOT",
        "/Users/macbookpro/AMIR_DATA/00_WORKS/01_PROJECTS/"
        "RAG-projects/RAG_finance",
    )).expanduser().resolve()
    report_dir = root / "reports/phase2/quality_improvement/NB22E.lab.v3"
    plan_path = report_dir / (
        "C040.R11_generation_plan."
        "608d2551794d2028f973e23a89e96079b0f8b2fead857f43c16079379f29b3bd.json"
    )
    plan_bytes = plan_path.read_bytes()
    plan_sha = sha(plan_bytes)
    require(
        plan_sha == "cda83957141dcfbf85e2073b9af21e65d8406da2e328da6c2b5f25a713579e85",
        "C042_C040_PLAN_HASH_MISMATCH",
    )
    pilot = dict(V3_R11_DEEPSEEK_PILOT)
    pilot_sha = pilot.pop("pilot_sha256")
    require(
        sha(encode(pilot)) == pilot_sha
        == "72b9e50483e1110d317da3d32635649230cea3666459abb24bad8018f818ec14"
        and pilot["source_plan_file_bytes_sha256"] == plan_sha,
        "C042_C041_PILOT_BINDING_MISMATCH",
    )

    env_path = root / ".env"
    require(env_path.is_file(), "C042_PROJECT_DOTENV_FILE_MISSING")
    config = dotenv_values(env_path, interpolate=False)
    base = str(config.get("ARVAN_GATEWAY_URL") or "").strip()
    key = str(config.get("ARVAN_GATEWAY_API_KEY") or "").strip()
    require(bool(base), "C042_ARVAN_GATEWAY_URL_MISSING")
    require(bool(key), "C042_ARVAN_GATEWAY_API_KEY_MISSING")

    try:
        url = urllib.parse.urlsplit(base)
        valid_url = (
            url.scheme == "https" and bool(url.hostname)
            and url.username is None and url.password is None
            and not url.query and not url.fragment
        )
    except ValueError:
        valid_url = False
    require(valid_url, "C042_INVALID_GATEWAY_HTTPS_URL")

    route = url.path.rstrip("/")
    if not route:
        route = "/v1/chat/completions"
    elif not route.endswith("/chat/completions"):
        route += "/chat/completions"
    endpoint = url._replace(path=route).geturl()
    endpoint_sha = sha(endpoint.encode("utf-8"))

    jobs = json.loads(plan_bytes)["jobs"]
    candidate_ids = [job["candidate_id"] for job in jobs]
    pilot_id = pilot["candidate_id"]
    require(
        len(jobs) == len(set(candidate_ids)) == 34 and pilot_id in candidate_ids,
        "C042_EXPECTED_34_DISTINCT_JOBS_WITH_PILOT",
    )

    prepared = []
    for job in sorted(jobs, key=lambda row: row["candidate_id"] != pilot_id):
        candidate_id = job["candidate_id"]
        messages = job["request"]["messages"]
        require(
            isinstance(messages, list) and bool(messages)
            and all(
                isinstance(message, dict)
                and isinstance(message.get("role"), str)
                and isinstance(message.get("content"), str)
                for message in messages
            ),
            "C042_INVALID_PLAN_MESSAGES",
        )
        payload = {
            "model": "DeepSeek-V4-Flash",
            "messages": messages,
            "temperature": 0,
            "max_tokens": 600,
            "thinking": {"type": "disabled"},
        }
        if candidate_id == pilot_id:
            require(payload == pilot["request"], "C042_PILOT_REQUEST_MISMATCH")

        body = encode(payload)
        request_id = sha(encode({
            "candidate_id": candidate_id,
            "plan_sha256": plan_sha,
            "endpoint_sha256": endpoint_sha,
            "request_sha256": sha(body),
        }))
        estimated_units = (
            (len(encode(messages)) + 1024) * input_rate
            + payload["max_tokens"] * output_rate
        )
        prepared.append((candidate_id, request_id, body, estimated_units))

    package_estimate = sum(row[3] for row in prepared)
    require(package_estimate < budget, "C042_PACKAGE_ESTIMATE_EXCEEDS_BUDGET")
    print(json.dumps({
        "cell": "C042",
        "planned_requests": len(prepared),
        "estimated_package_toman": package_estimate / 1_000_000,
        "budget_toman": C042_BUDGET_TOMAN,
        "rate_basis": "CONFIGURED_REFERENCE_RATES",
    }, ensure_ascii=False), flush=True)

    receipt_dir = report_dir / "C042.R11_deepseek_receipts"
    receipt_dir.mkdir(exist_ok=True)

    class NoRedirect(urllib.request.HTTPRedirectHandler):
        def redirect_request(self, req, fp, code, msg, headers, newurl):
            return None

    opener = urllib.request.build_opener(NoRedirect())
    allowed_models = {"deepseek-v4-flash", "deepseek-v4-flash-0731"}
    completed = []
    attempted = 0
    observed_units = 0
    blocker = None

    for index, (candidate_id, request_id, body, _) in enumerate(prepared):
        identity = {"request_id": request_id, "candidate_id": candidate_id}
        started_path = receipt_dir / f"{request_id}.started.json"
        response_path = receipt_dir / f"{request_id}.response.json"

        if response_path.exists():
            try:
                receipt = json.loads(response_path.read_bytes())
            except (ValueError, OSError):
                blocker = "C042_EXISTING_RECEIPT_UNREADABLE"
                break
        else:
            if started_path.exists():
                blocker = "C042_PREVIOUS_ATTEMPT_WITHOUT_RESPONSE_RECEIPT"
                break
            if observed_units + sum(row[3] for row in prepared[index:]) >= budget:
                blocker = "C042_REMAINING_ESTIMATE_EXCEEDS_BUDGET"
                break

            request = urllib.request.Request(
                endpoint,
                data=body,
                headers={
                    "Authorization": f"apikey {key}",
                    "Content-Type": "application/json",
                },
                method="POST",
            )
            try:
                write_once(started_path, identity)
            except FileExistsError:
                blocker = "C042_ATTEMPT_ALREADY_CLAIMED"
                break

            attempted += 1
            try:
                with opener.open(request, timeout=180) as response:
                    require(response.status == 200, "C042_UNEXPECTED_HTTP_STATUS")
                    raw = response.read()
                receipt = {
                    **identity,
                    "status": "HTTP_200_RECEIVED",
                    "response_body": raw.decode("utf-8", errors="replace"),
                    "response_bytes_sha256": sha(raw),
                }
            except Exception as error:
                receipt = {
                    **identity,
                    "status": "SEND_OR_READ_ERROR",
                    "error_type": type(error).__name__,
                    "http_status": getattr(error, "code", None),
                }
            write_once(response_path, receipt)

        try:
            require(
                all(receipt.get(field) == value for field, value in identity.items()),
                "C042_RECEIPT_IDENTITY_MISMATCH",
            )
            if receipt["status"] != "HTTP_200_RECEIVED":
                http_status = receipt.get("http_status")
                blocker = (
                    f"C042_HTTP_{http_status}"
                    if type(http_status) is int else "C042_SEND_OR_READ_ERROR"
                )
                break

            response_body = receipt["response_body"]
            require(
                sha(response_body.encode("utf-8")) == receipt["response_bytes_sha256"],
                "C042_RESPONSE_BYTES_MISMATCH",
            )
            envelope = json.loads(response_body)
            usage = envelope["usage"]
            prompt_tokens = usage["prompt_tokens"]
            completion_tokens = usage["completion_tokens"]
            require(
                type(prompt_tokens) is int and prompt_tokens >= 0
                and type(completion_tokens) is int and completion_tokens >= 0,
                "C042_INVALID_USAGE",
            )
            observed_units += (
                prompt_tokens * input_rate + completion_tokens * output_rate
            )
            actual_model = envelope["model"].strip().rsplit("/", 1)[-1].lower()
            choice = envelope["choices"][0]
            finish_reason = choice["finish_reason"]
            content = choice["message"]["content"]
            require(actual_model in allowed_models, "C042_UNEXPECTED_RESPONSE_MODEL")
            require(finish_reason == "stop", "C042_RESPONSE_NOT_FINISHED")
            require(isinstance(content, str) and content.strip(), "C042_EMPTY_RESPONSE")
        except (KeyError, IndexError, TypeError, AttributeError, ValueError) as error:
            code = str(error)
            blocker = code if code.startswith("C042_") else "C042_INVALID_RESPONSE_ENVELOPE"
            break

        completed.append({
            "candidate_id": candidate_id,
            "receipt_path": str(response_path.relative_to(root)),
            "response_bytes_sha256": receipt["response_bytes_sha256"],
            "response_model": actual_model,
            "finish_reason": finish_reason,
            "prompt_tokens": prompt_tokens,
            "completion_tokens": completion_tokens,
        })
        print(f"C042 {len(completed)}/{len(prepared)}", flush=True)
        if observed_units >= budget:
            blocker = "C042_OBSERVED_USAGE_REACHED_BUDGET"
            break

    result = {
        "cell": "C042",
        "roadmap_action": "R11",
        "status": "BATCH_PAUSED" if blocker else "BATCH_RESPONSES_RECEIVED",
        "planned_requests": len(prepared),
        "responses_received_and_checked": len(completed),
        "new_api_calls_attempted": attempted,
        "estimated_package_toman": package_estimate / 1_000_000,
        "cost_estimated_from_read_usage_toman": observed_units / 1_000_000,
        "input_toman_per_million": input_rate,
        "output_toman_per_million": output_rate,
        "blocker": blocker,
        "blocked_candidate_id": candidate_id if blocker else None,
        "receipts": completed,
        "selected_version_gate": "BLOCKED",
    }
    output_bytes = encode(result)
    output_sha = sha(output_bytes)
    output_path = report_dir / f"C042.R11_deepseek_batch.{output_sha}.json"
    try:
        write_once(output_path, result)
    except FileExistsError:
        pass
    require(output_path.read_bytes() == output_bytes, "C042_SUMMARY_RECEIPT_CONFLICT")

    print(json.dumps({
        **{field: value for field, value in result.items() if field != "receipts"},
        "output": {
            "path": str(output_path.relative_to(root)),
            "file_bytes_sha256": output_sha,
            "reload_validation": "PASS",
        },
    }, ensure_ascii=False, indent=2))
    return result


V3_C042 = run_v3_c042()
````

### C043 — اتصال پاسخ‌های DeepSeek به ورودی بازبینی کار ۱۱

خروجی C042، برنامهٔ C040 و تعریف‌های C037 با هش ثبت‌شده خوانده می‌شوند.
برای هر یک از ۳۴ پرونده، هویت درخواست، رسید پاسخ و هش محتوای دریافتی
کنترل می‌شود. پیام‌های اصلی درخواست، پاسخ و تعریف‌های ۳۶ سناریو در
یک بستهٔ آفلاین جمع می‌شوند.

این سل هیچ درخواست تازه‌ای ارسال نمی‌کند. نتیجهٔ محتوایی سناریوها

````python
# C044 | R11 | Run after C001.
# C001 provides Path, os, json, hashlib, and tempfile.

def run_v3_c044():
    def require(condition, code):
        if not condition:
            raise ValueError(code)

    def sha(data):
        return hashlib.sha256(data).hexdigest()

    def encode(value):
        return json.dumps(
            value, ensure_ascii=False, sort_keys=True,
            separators=(",", ":"), allow_nan=False,
        ).encode("utf-8")

    root = Path(os.environ.get(
        "RAG_FINANCE_ROOT",
        "/Users/macbookpro/AMIR_DATA/00_WORKS/01_PROJECTS/"
        "RAG-projects/RAG_finance",
    )).expanduser().resolve()
    report_dir = root / "reports/phase2/quality_improvement/NB22E.lab.v3"
    source_sha = "811af55fabd63d695bd2767c762055c4997d81e4acd7c36f7bdde717421e6145"
    source_path = report_dir / f"C043.R11_scenario_review_inputs.{source_sha}.json"

    source_bytes = source_path.read_bytes()
    require(sha(source_bytes) == source_sha, "C044_C043_HASH_MISMATCH")
    source = json.loads(source_bytes)
    require(
        source["cell"] == "C043"
        and source["roadmap_action"] == "R11"
        and source["status"] == "DIAGNOSTIC_RESPONSES_PACKAGED_FOR_SCENARIO_REVIEW"
        and source["scope"] == "FROZEN_CHAR_FUSION_ON_HISTORICAL_POOL50_INCUMBENT",
        "C044_C043_SCOPE_MISMATCH",
    )

    responses = source["responses"]
    by_candidate = {row["candidate_id"]: row for row in responses}
    scenarios = source["scenario_definitions"]["selected_version_execution_requests"]
    scenario_ids = {row["regression_spec_id"] for row in scenarios}
    require(len(responses) == len(by_candidate) == 34, "C044_RESPONSE_SET_MISMATCH")
    require(len(scenarios) == len(scenario_ids) == 36, "C044_SCENARIO_SET_MISMATCH")
    require(
        {row["candidate_id"] for row in scenarios} == set(by_candidate),
        "C044_SCENARIO_RESPONSE_COVERAGE_MISMATCH",
    )

    kind_counts = {"ANSWER_DECISION": 0, "CITATION_WITNESS": 0}
    observation_counts = {}
    dispositions = []
    abstention_prefix = "در منابع موجود پاسخ کافی پیدا نشد"

    for scenario in scenarios:
        candidate_id = scenario["candidate_id"]
        response = by_candidate[candidate_id]
        content = response["response_content"]
        require(
            isinstance(content, str) and bool(content.strip()),
            "C044_EMPTY_RESPONSE",
        )

        kind = scenario["kind"]
        require(kind in kind_counts, "C044_UNEXPECTED_SCENARIO_KIND")
        kind_counts[kind] += 1
        row = {
            "regression_spec_id": scenario["regression_spec_id"],
            "candidate_id": candidate_id,
            "kind": kind,
            "response_bytes_sha256": response["response_bytes_sha256"],
            "response_content_sha256": sha(content.encode("utf-8")),
            "generation_input_sha256": response["generation_input_sha256"],
            "required_execution_evidence": scenario["required_execution_evidence"],
            "diagnostic_semantic_verdict": None,
            "product_regression_status": "BLOCKED",
        }

        if kind == "ANSWER_DECISION":
            expectation = scenario["behavior_expectation"]
            expected_behavior = expectation["expected_behavior"]
            first_line = content.strip().splitlines()[0].strip()
            explicit_abstention = first_line.startswith(abstention_prefix)

            if explicit_abstention and expected_behavior in {
                "ANSWER_SUPPORTED", "BOUNDED_ANSWER_SUPPORTED",
            }:
                observation = "EXPLICIT_ABSTENTION_VS_HISTORICAL_ANSWER_EXPECTATION"
            elif explicit_abstention and expected_behavior == "CLARIFICATION_REQUIRED":
                observation = "EXPLICIT_ABSTENTION_VS_HISTORICAL_CLARIFICATION"
            elif expected_behavior == "CLARIFICATION_REQUIRED":
                observation = "CHECK_IF_RESPONSE_REQUESTS_REQUIRED_CLARIFICATION"
            else:
                observation = "CHECK_ANSWER_SUPPORT_AND_REQUIRED_LIMIT"

            row.update({
                "historical_behavior_expectation": expectation,
                "explicit_abstention_prefix_detected": explicit_abstention,
                "opening_line": first_line,
                "next_review": "COMPARE_QUESTION_SUPPLIED_EVIDENCE_AND_RESPONSE",
            })
        else:
            observation = "HISTORICAL_LINK_REQUIRES_BINDING_TO_NEW_RESPONSE"
            row.update({
                "historical_link_expectation": scenario["historical_link_expectation"],
                "next_review": "LOCATE_NEW_CLAIM_AND_CHECK_EXACT_SUPPORTING_WITNESS",
            })

        row["observation"] = observation
        observation_counts[observation] = observation_counts.get(observation, 0) + 1
        dispositions.append(row)

    require(
        kind_counts == {"ANSWER_DECISION": 31, "CITATION_WITNESS": 5},
        "C044_SCENARIO_KIND_COUNTS_MISMATCH",
    )

    result = {
        "cell": "C044",
        "roadmap_action": "R11",
        "status": "ALL_36_SCENARIOS_SCREENED_FOR_DIAGNOSTIC_REVIEW",
        "scope": source["scope"],
        "input": {
            "path": str(source_path.relative_to(root)),
            "file_bytes_sha256": source_sha,
        },
        "distinct_responses_accounted": len(by_candidate),
        "scenario_counts": kind_counts,
        "observation_counts": observation_counts,
        "dispositions": dispositions,
        "semantic_scenario_verdicts_recorded": 0,
        "product_regression_tests_executed": 0,
        "new_api_calls": 0,
        "selected_version_gate": "BLOCKED",
        "all_R11_items_complete": False,
    }

    output_bytes = encode(result)
    output_sha = sha(output_bytes)
    output_path = report_dir / f"C044.R11_diagnostic_scenario_screen.{output_sha}.json"
    if not output_path.exists():
        with tempfile.NamedTemporaryFile(
            dir=report_dir, prefix="C044.", suffix=".tmp", delete=False,
        ) as handle:
            temporary_path = Path(handle.name)
            try:
                handle.write(output_bytes)
                handle.flush()
                os.fsync(handle.fileno())
            finally:
                handle.close()
        try:
            os.replace(temporary_path, output_path)
        finally:
            temporary_path.unlink(missing_ok=True)

    require(output_path.read_bytes() == output_bytes, "C044_OUTPUT_RELOAD_MISMATCH")
    print(json.dumps({
        "cell": result["cell"],
        "status": result["status"],
        "distinct_responses_accounted": result["distinct_responses_accounted"],
        "scenario_counts": kind_counts,
        "observation_counts": observation_counts,
        "semantic_scenario_verdicts_recorded": 0,
        "new_api_calls": 0,
        "selected_version_gate": "BLOCKED",
        "output": {
            "path": str(output_path.relative_to(root)),
            "file_bytes_sha256": output_sha,
            "reload_validation": "PASS",
        },
    }, ensure_ascii=False, indent=2))
    return result


V3_C044 = run_v3_c044()
````

````python
# C046 | R11 | Run after C001.
# C001 provides Path, os, json, hashlib, re, and tempfile.

def run_v3_c046():
    def sha(data):
        return hashlib.sha256(data).hexdigest()

    def require(condition, code):
        if not condition:
            raise ValueError(code)

    root = Path(os.environ.get(
        "RAG_FINANCE_ROOT",
        "/Users/macbookpro/AMIR_DATA/00_WORKS/01_PROJECTS/"
        "RAG-projects/RAG_finance",
    )).expanduser().resolve()
    folder = root / "reports/phase2/quality_improvement/NB22E.lab.v3"

    c043_sha = "811af55fabd63d695bd2767c762055c4997d81e4acd7c36f7bdde717421e6145"
    c045_sha = "3e76c715fecbb40cf72f84463c60effa7afa7e7779f751ff8d334b252603fbb7"

    def read_pinned(stem, digest):
        path = folder / f"{stem}.{digest}.json"
        raw = path.read_bytes()
        require(sha(raw) == digest, f"C046_INPUT_HASH_MISMATCH: {stem}")
        return json.loads(raw)

    source = read_pinned("C043.R11_scenario_review_inputs", c043_sha)
    ledger = read_pinned("C045.R11_clarification_behavior_review", c045_sha)
    require(
        ledger["input_file_hashes"]["C043"] == c043_sha
        and ledger["scope"] == source["scope"],
        "C046_SOURCE_BINDING_MISMATCH",
    )

    scenarios = source["scenario_definitions"]["selected_version_execution_requests"]
    citations = {
        item["regression_spec_id"]: item
        for item in scenarios if item["kind"] == "CITATION_WITNESS"
    }
    responses = {item["candidate_id"]: item for item in source["responses"]}
    require(
        len(responses) == len(source["responses"]) == 34
        and len(scenarios) == len({item["regression_spec_id"] for item in scenarios}) == 36
        and len(citations) == 5
        and len(ledger["dispositions"]) == 36
        and {row["regression_spec_id"] for row in ledger["dispositions"]}
        == {item["regression_spec_id"] for item in scenarios},
        "C046_SCENARIO_SET_MISMATCH",
    )

    # Preserve the source body exactly apart from the prompt's block separator.
    header = re.compile(
        r"(?m)^\[منبع \d+\]\nشناسه چانک: ([^\n]+)\n\nمتن:\n"
    )
    results = []
    counts = {}

    for previous in ledger["dispositions"]:
        row = dict(previous)
        scenario_id = row["regression_spec_id"]
        scenario = citations.get(scenario_id)
        if scenario is None:
            results.append(row)
            continue

        candidate = scenario["candidate_id"]
        response = responses[candidate]
        require(
            row["candidate_id"] == candidate
            and row["response_bytes_sha256"] == response["response_bytes_sha256"]
            and row["historical_link_expectation"] == scenario["historical_link_expectation"],
            f"C046_CITATION_BINDING_MISMATCH: {scenario_id}",
        )

        link = scenario["historical_link_expectation"]
        evidence_id = link["evidence_id"]
        expected_hash = link["evidence_text_sha256"]
        prompt = response["messages_used_by_c042"][-1]["content"]
        headers = list(header.finditer(prompt))
        matched_bodies = []

        for index, match in enumerate(headers):
            if match.group(1).strip() != evidence_id:
                continue
            end = headers[index + 1].start() if index + 1 < len(headers) else len(prompt)
            block = prompt[match.end():end]
            # Accept a boundary adjustment only when the historical text hash verifies it.
            variants = {block, block.removesuffix("\n"), block.removesuffix("\n\n")}
            matched_bodies.append(variants)

        exact_texts = {
            body for variants in matched_bodies for body in variants
            if sha(body.encode("utf-8")) == expected_hash
        }
        span_checks = []
        if not headers:
            status = "PROMPT_SOURCE_FORMAT_UNRECOGNIZED"
        elif not matched_bodies:
            status = "HISTORICAL_SOURCE_ID_NOT_FOUND_IN_PARSED_INPUT"
        elif not exact_texts:
            status = "HISTORICAL_SOURCE_TEXT_HASH_NOT_VERIFIED"
        else:
            require(len(exact_texts) == 1, "C046_MULTIPLE_EXACT_SOURCE_TEXTS")
            body = next(iter(exact_texts))
            spans = link.get("exact_spans", [])
            if not spans:
                status = "HISTORICAL_EXACT_SPAN_NOT_RECORDED"
            elif link.get("span_unit") != "UNICODE_CODE_POINT":
                status = "HISTORICAL_SPAN_UNIT_UNVERIFIED"
            else:
                for span in spans:
                    start, end = span["start"], span["end"]
                    quote = span["quote"]
                    exact_match = (
                        type(start) is int and type(end) is int
                        and 0 <= start < end <= len(body)
                        and isinstance(quote, str)
                        and body[start:end] == quote
                    )
                    span_checks.append({
                        "start": start, "end": end, "quote": quote,
                        "exact_match": exact_match,
                    })
                status = (
                    "HISTORICAL_SPANS_VERIFIED_IN_GENERATION_INPUT"
                    if all(span["exact_match"] for span in span_checks)
                    else "HISTORICAL_SPAN_MISMATCH"
                )

        row["historical_witness_input_check"] = {
            "status": status,
            "evidence_id": evidence_id,
            "expected_evidence_text_sha256": expected_hash,
            "parsed_source_count": len(headers),
            "matching_source_id_count": len(matched_bodies),
            "historical_text_sha256_verified": bool(exact_texts),
            "span_checks": span_checks,
            "new_response_claim_support_established": False,
        }
        counts[status] = counts.get(status, 0) + 1
        results.append(row)

    require(sum(counts.values()) == 5, "C046_CITATION_COUNT_MISMATCH")
    result = {
        "cell": "C046",
        "roadmap_action": "R11",
        "status": "FIVE_HISTORICAL_WITNESS_INPUT_CHECKS_RECORDED",
        "scope": source["scope"],
        "input_file_hashes": {"C043": c043_sha, "C045": c045_sha},
        "scenario_count": len(results),
        "citation_input_check_counts": counts,
        "answer_decision_scenarios_pending_content_review": 26,
        "new_api_calls": 0,
        "product_regression_tests_executed": 0,
        "selected_version_gate": "BLOCKED",
        "dispositions": results,
    }
    data = json.dumps(
        result, ensure_ascii=False, sort_keys=True, separators=(",", ":"),
        allow_nan=False,
    ).encode("utf-8")
    digest = sha(data)
    path = folder / f"C046.R11_citation_witness_input_checks.{digest}.json"

    if not path.exists():
        with tempfile.NamedTemporaryFile(dir=folder, prefix="C046.", delete=False) as handle:
            temporary = Path(handle.name)
            try:
                handle.write(data)
                handle.flush()
                os.fsync(handle.fileno())
            except BaseException:
                temporary.unlink(missing_ok=True)
                raise
        try:
            os.replace(temporary, path)
        finally:
            temporary.unlink(missing_ok=True)

    require(path.read_bytes() == data, "C046_OUTPUT_RELOAD_MISMATCH")
    print(json.dumps({
        **{key: value for key, value in result.items() if key != "dispositions"},
        "output": {
            "path": str(path.relative_to(root)),
            "file_bytes_sha256": digest,
            "reload_validation": "PASS",
        },
    }, ensure_ascii=False, indent=2))
    return result


V3_C046 = run_v3_c046()
````

پشتیبانی می‌کند. برای فقر کارگر، پاسخ به مثال و تحلیل منبع مقید است.
برای ازدواج، پاسخ به بُعد مالی محدود است.

نقل‌قول، بازهٔ دقیق، شناسهٔ قطعه و هش پاسخ در دفتر C064 ثبت می‌شوند.
اجرای سل آفلاین است و Gate محصول همچنان BLOCKED است.

````python
# C065 | R11.3 | C001 provides Path, os, json, hashlib, tempfile.

def run_v3_c065():
    def sha(raw):
        return hashlib.sha256(raw).hexdigest()

    def require(condition, code):
        if not condition:
            raise ValueError(code)

    root = Path(os.environ.get(
        "RAG_FINANCE_ROOT",
        "/Users/macbookpro/AMIR_DATA/00_WORKS/01_PROJECTS/"
        "RAG-projects/RAG_finance",
    )).expanduser().resolve()
    folder = root / "reports/phase2/quality_improvement/NB22E.lab.v3"
    source_sha = "811af55fabd63d695bd2767c762055c4997d81e4acd7c36f7bdde717421e6145"
    ledger_sha = "86094962b7af2a189c9ad83ffd3375bbc28bddfae8d86483e8f8121ec37fe319"

    def read_pinned(stem, digest):
        raw = (folder / f"{stem}.{digest}.json").read_bytes()
        require(sha(raw) == digest, f"C065_INPUT_HASH_MISMATCH: {stem}")
        return json.loads(raw)

    source = read_pinned("C043.R11_scenario_review_inputs", source_sha)
    ledger = read_pinned("C064.R11_investment_objective_witness_review", ledger_sha)
    require(
        ledger["input_file_hashes"]["C043"] == source_sha
        and ledger["scope"] == source["scope"],
        "C065_LEDGER_BINDING_MISMATCH",
    )

    rows = [dict(row) for row in ledger["dispositions"]]
    by_id = {row["regression_spec_id"]: row for row in rows}
    responses = {row["candidate_id"]: row for row in source["responses"]}
    require(
        len(rows) == len(by_id) == 36
        and len(source["responses"]) == len(responses) == 34,
        "C065_POPULATION_MISMATCH",
    )

    batch = [
        row for row in rows
        if row.get("diagnostic_question_response_review", {}).get("scope")
        == "QUESTION_AND_FULL_RESPONSE_CONTENT"
    ]
    unreviewed = {
        row["regression_spec_id"] for row in batch
        if "diagnostic_evidence_sufficiency_review" not in row
    }
    reviews = {
        "REG_73280d36a71f841b95639458": {
            "question": "از فقر کارگر چه می‌شود؟",
            "quote": (
                "این یعنی سرمایه‌گذار همواره کمتر از ارزش کاری کارگر پرداخت "
                "می‌کند. در واقع، از فقر کارگر، ثروتمند می‌شود."
            ),
            "bounded_answer_fa": (
                "در مثال و تحلیل منبع، سرمایه‌گذار از پرداخت کمتر از ارزش کار "
                "بهره می‌برد و ثروتمند می‌شود."
            ),
        },
        "REG_c05a077fb1b2f216431f2cde": {
            "question": "جزء اصلی فرآیند ازدواج چیست؟",
            "quote": "ابعاد ازدواج طبعاً همون همکاری مالیه.",
            "bounded_answer_fa": "در بحث بُعد مالی ازدواج، منبع به همکاری مالی اشاره می‌کند.",
        },
    }
    require(
        len(batch) == ledger["batch_scenarios_pending_final_evidence_sufficiency"] == 9
        and len(unreviewed) == ledger["batch_scenarios_without_evidence_review"] == 8
        and set(reviews) <= unreviewed,
        "C065_REVIEW_BATCH_MISMATCH",
    )

    for scenario_id, review in reviews.items():
        row = by_id[scenario_id]
        response = responses[row["candidate_id"]]
        content = response["response_content"]
        prompt = response["messages_used_by_c042"][-1]["content"]
        question, separator, _ = prompt.partition("\nمنابع:")
        require(
            bool(separator)
            and question.strip() == f"سؤال:\n{review['question']}"
            and row["kind"] == "ANSWER_DECISION"
            and row["diagnostic_question_response_review"]["status"] == "BARE_ABSTENTION"
            and content.strip() == "در منابع موجود پاسخ کافی پیدا نشد."
            and sha(content.encode("utf-8")) == row["response_content_sha256"]
            and response["response_bytes_sha256"] == row["response_bytes_sha256"]
            and response["generation_input_sha256"] == row["generation_input_sha256"],
            f"C065_RESPONSE_BINDING_MISMATCH: {scenario_id}",
        )

        _, marker, tail = prompt.partition("\n[منبع 1]\n")
        block, next_marker, _ = tail.partition("\n[منبع 2]\n")
        header, text_marker, body = block.partition("\n\nمتن:\n")
        prefix = "شناسه چانک: "
        evidence_id = header.removeprefix(prefix).strip()
        quote = review["quote"]
        require(
            bool(marker and next_marker and text_marker)
            and header.startswith(prefix)
            and bool(evidence_id) and "\n" not in evidence_id
            and body.count(quote) == 1,
            f"C065_EXACT_WITNESS_MISMATCH: {scenario_id}",
        )

        start = body.index(quote)
        row["diagnostic_evidence_sufficiency_review"] = {
            "status": "SUPPLIED_EVIDENCE_SUFFICIENT_FOR_BOUNDED_ANSWER",
            "final_scenario_verdict": "BARE_ABSTENTION_MISSED_SUPPORTED_BOUNDED_ANSWER",
            "verdict_scope": "DIAGNOSTIC_RESPONSE_AND_SUPPLIED_EVIDENCE",
            "question": review["question"],
            "bounded_answer_fa": review["bounded_answer_fa"],
            "full_supplied_evidence_sufficiency_assessed": False,
            "response_content_sha256": row["response_content_sha256"],
            "generation_input_sha256": row["generation_input_sha256"],
            "witness": {
                "source_number": 1,
                "evidence_id": evidence_id,
                "rendered_source_body_sha256": sha(body.encode("utf-8")),
                "offset_reference": "C043_RENDERED_SOURCE_BODY",
                "span_unit": "UNICODE_CODE_POINT",
                "start": start,
                "end": start + len(quote),
                "quote": quote,
            },
        }

    result = {
        "cell": "C065",
        "roadmap_action": "R11",
        "roadmap_subtask": "R11.3",
        "status": "TWO_MISSED_BOUNDED_ANSWERS_RECORDED",
        "scope": ledger["scope"],
        "input_file_hashes": {"C043": source_sha, "C064": ledger_sha},
        "reviewed_regression_spec_ids": list(reviews),
        "answer_decisions_pending_initial_response_review": 0,
        "batch_scenarios_without_evidence_review": sum(
            "diagnostic_evidence_sufficiency_review" not in row for row in batch
        ),
        "batch_scenarios_pending_final_evidence_sufficiency": sum(
            row.get("diagnostic_evidence_sufficiency_review", {})
            .get("final_scenario_verdict") is None
            for row in batch
        ),
        "citation_scenarios_pending_full_current_response_review": (
            ledger["citation_scenarios_pending_full_current_response_review"]
        ),
        "new_api_calls": 0,
        "product_regression_tests_executed": 0,
        "selected_version_gate": "BLOCKED",
        "dispositions": rows,
    }
    data = json.dumps(
        result, ensure_ascii=False, sort_keys=True, separators=(",", ":"),
        allow_nan=False,
    ).encode("utf-8")
    digest = sha(data)
    output = folder / f"C065.R11_missed_bounded_answers_review.{digest}.json"

    if not output.exists():
        temporary = None
        try:
            with tempfile.NamedTemporaryFile(
                dir=folder, prefix="C065.", delete=False,
            ) as handle:
                temporary = Path(handle.name)
                handle.write(data)
                handle.flush()
                os.fsync(handle.fileno())
            os.replace(temporary, output)
        finally:
            if temporary is not None:
                temporary.unlink(missing_ok=True)

    require(output.read_bytes() == data, "C065_OUTPUT_RELOAD_MISMATCH")
    print(json.dumps({
        **{key: value for key, value in result.items() if key != "dispositions"},
        "output": {
            "path": str(output.relative_to(root)),
            "file_bytes_sha256": digest,
            "reload_validation": "PASS",
        },
    }, ensure_ascii=False, indent=2))
    return result


V3_C065 = run_v3_c065()
````

### C081 — یافتن رسیدهای اجرای نسخهٔ منتخب در R11.3

````python
# C081 | R11.3 | C001 provides Path, os, json, hashlib, tempfile.

def run_v3_c081():
    def sha(raw):
        return hashlib.sha256(raw).hexdigest()

    root = Path(os.environ.get(
        "RAG_FINANCE_ROOT",
        "/Users/macbookpro/AMIR_DATA/00_WORKS/01_PROJECTS/"
        "RAG-projects/RAG_finance",
    )).expanduser().resolve()
    folder = root / "reports/phase2/quality_improvement/NB22E.lab.v3"
    c080_sha = "15dba7dddba32ec40bd63547637735794932a72c02a970df4e45b1f49678857b"
    c080_path = folder / (
        f"C080.R11_diagnostic_residuals_and_execution_requirements.{c080_sha}.json"
    )
    raw = c080_path.read_bytes()
    if sha(raw) != c080_sha:
        raise ValueError("C081_C080_HASH_MISMATCH")

    ledger = json.loads(raw)
    scenarios = ledger["product_execution_requirements"]
    candidate_ids = {row["candidate_id"] for row in scenarios}
    if (
        ledger["cell"] != "C080"
        or ledger["next_product_dependency"]
        != "VERIFY_SELECTED_VARIANT_RUNTIME_AND_MATCHED_OUTPUT_RECEIPTS"
        or len(scenarios) != 36
        or len({row["regression_spec_id"] for row in scenarios}) != 36
        or len(candidate_ids) != 34
    ):
        raise ValueError("C081_C080_SCENARIO_BINDING_MISMATCH")

    # Explicit local search scope. Only JSON and JSONL are inspected.
    search_roots = ("reports/phase2", "artifacts/phase2", "data/phase2")
    size_limit = 64 * 1024 * 1024
    id_bytes = {candidate: candidate.encode("utf-8") for candidate in candidate_ids}
    diagnostic_receipts = folder / "C042.R11_deepseek_receipts"
    candidates = []
    skipped = []
    missing_roots = []
    scanned_count = 0

    for relative_root in search_roots:
        base = root / relative_root
        if not base.is_dir():
            missing_roots.append(relative_root)
            continue

        for path in sorted(base.rglob("*")):
            if path.is_symlink() or path.suffix.lower() not in {".json", ".jsonl"}:
                continue
            if not path.is_file() or not path.resolve().is_relative_to(root):
                continue
            if diagnostic_receipts in path.parents or (
                path.parent == folder and ".R11_" in path.name
            ):
                continue

            relative_path = str(path.relative_to(root))
            try:
                if path.stat().st_size > size_limit:
                    skipped.append({"path": relative_path, "reason": "SIZE_LIMIT"})
                    continue
                with path.open("rb") as handle:
                    payload = handle.read(size_limit + 1)
                if len(payload) > size_limit:
                    skipped.append({"path": relative_path, "reason": "SIZE_LIMIT"})
                    continue
            except OSError as error:
                skipped.append({"path": relative_path, "reason": type(error).__name__})
                continue

            scanned_count += 1
            mentions_variant = b"char_rrf" in payload
            if not mentions_variant:
                continue

            matched_ids = sorted(
                candidate for candidate, token in id_bytes.items() if token in payload
            )
            name = path.name.lower()
            looks_like_receipt = any(
                term in name for term in ("receipt", "runtime", "execution", "manifest")
            )
            if matched_ids or looks_like_receipt:
                candidates.append({
                    "path": relative_path,
                    "file_bytes_sha256": sha(payload),
                    "size_bytes": len(payload),
                    "candidate_ids_found_as_text": matched_ids,
                    "receipt_related_filename": looks_like_receipt,
                    "runtime_binding_verified": False,
                })

    candidates.sort(key=lambda row: (
        -len(row["candidate_ids_found_as_text"]), row["path"],
    ))
    matched_candidate_ids = {
        candidate for row in candidates for candidate in row["candidate_ids_found_as_text"]
    }

    result = {
        "cell": "C081",
        "roadmap_action": "R11",
        "roadmap_subtask": "R11.3",
        "status": "LOCAL_RUNTIME_RECEIPT_CANDIDATES_INVENTORIED",
        "input_file_hashes": {"C080": c080_sha},
        "selected_variant_searched": "char_rrf",
        "scenario_count": len(scenarios),
        "distinct_candidates_expected": len(candidate_ids),
        "search_roots": list(search_roots),
        "missing_search_roots": missing_roots,
        "scanned_file_count": scanned_count,
        "skipped_file_count": len(skipped),
        "skipped_files": skipped,
        "candidate_file_count": len(candidates),
        "distinct_candidate_ids_found_as_text": len(matched_candidate_ids),
        "candidate_files": candidates,
        "verified_product_runtime_receipts_in_this_cell": 0,
        "product_regression_tests_executed": 0,
        "new_api_calls": 0,
        "selected_version_gate": "BLOCKED",
    }

    data = json.dumps(
        result, ensure_ascii=False, sort_keys=True, separators=(",", ":"),
        allow_nan=False,
    ).encode("utf-8")
    digest = sha(data)
    output = folder / f"C081.R11_local_runtime_receipt_inventory.{digest}.json"
    if not output.exists():
        temporary = None
        try:
            with tempfile.NamedTemporaryFile(
                dir=folder, prefix="C081.", delete=False,
            ) as handle:
                temporary = Path(handle.name)
                handle.write(data)
                handle.flush()
                os.fsync(handle.fileno())
            os.replace(temporary, output)
        finally:
            if temporary is not None:
                temporary.unlink(missing_ok=True)

    if output.read_bytes() != data:
        raise ValueError("C081_OUTPUT_RELOAD_MISMATCH")

    print(json.dumps({
        **{key: value for key, value in result.items()
           if key not in {"candidate_files", "skipped_files"}},
        "first_five_candidate_files": [
            {
                "path": row["path"],
                "matched_candidate_count": len(row["candidate_ids_found_as_text"]),
            }
            for row in candidates[:5]
        ],
        "output": {
            "path": str(output.relative_to(root)),
            "file_bytes_sha256": digest,
            "reload_validation": "PASS",
        },
    }, ensure_ascii=False, indent=2))
    return result


V3_C081 = run_v3_c081()
````

### C082 — بررسی فایل‌های جاافتاده از جست‌وجوی رسید در R11.3

می‌شوند. فایل‌های بزرگ به‌صورت تکه‌ای بررسی می‌شوند. اشاره به نسخهٔ
char_rrf و شناسهٔ پرونده‌ها در هر فایل جدا ثبت می‌شود تا سرنخ‌های
احتمالی برای بررسی ساختار رسید مشخص شوند.

این سل آفلاین است و Gate نسخهٔ منتخب BLOCKED می‌ماند.

</div>

````python
# C082 | R11.3 | C001 provides Path, os, json, hashlib, tempfile.

def run_v3_c082():
    def sha(raw):
        return hashlib.sha256(raw).hexdigest()

    def require(condition, code):
        if not condition:
            raise ValueError(code)

    root = Path(os.environ.get(
        "RAG_FINANCE_ROOT",
        "/Users/macbookpro/AMIR_DATA/00_WORKS/01_PROJECTS/"
        "RAG-projects/RAG_finance",
    )).expanduser().resolve()
    folder = root / "reports/phase2/quality_improvement/NB22E.lab.v3"
    c080_sha = "15dba7dddba32ec40bd63547637735794932a72c02a970df4e45b1f49678857b"
    c081_sha = "8305bf10170a6e010c2e9949a903d9efb4174c87b03095d3831dbd47dbb0e02e"

    def read_pinned(stem, digest):
        raw = (folder / f"{stem}.{digest}.json").read_bytes()
        require(sha(raw) == digest, f"C082_INPUT_HASH_MISMATCH: {stem}")
        return json.loads(raw)

    requirements = read_pinned(
        "C080.R11_diagnostic_residuals_and_execution_requirements", c080_sha,
    )
    inventory = read_pinned("C081.R11_local_runtime_receipt_inventory", c081_sha)
    skipped = inventory["skipped_files"]
    scenarios = requirements["product_execution_requirements"]
    candidate_ids = {row["candidate_id"] for row in scenarios}
    require(
        inventory["cell"] == "C081"
        and inventory["input_file_hashes"]["C080"] == c080_sha
        and inventory["selected_variant_searched"] == "char_rrf"
        and inventory["candidate_file_count"] == 0
        and len(skipped) == inventory["skipped_file_count"] == 33
        and len({row["path"] for row in skipped}) == 33
        and len(scenarios) == 36
        and len(candidate_ids) == 34,
        "C082_INPUT_BINDING_MISMATCH",
    )

    tokens = {candidate.encode("utf-8"): candidate for candidate in candidate_ids}
    variant_token = b"char_rrf"
    overlap = max(len(token) for token in (*tokens, variant_token)) - 1
    allowed_roots = [(root / item).resolve() for item in inventory["search_roots"]]
    files = []

    for index, skipped_row in enumerate(skipped, start=1):
        relative = Path(skipped_row["path"])
        path = root / relative
        resolved = path.resolve()
        require(
            not relative.is_absolute()
            and ".." not in relative.parts
            and resolved.is_relative_to(root)
            and any(resolved.is_relative_to(base) for base in allowed_roots)
            and path.suffix.lower() in {".json", ".jsonl"},
            "C082_PATH_OUTSIDE_DECLARED_SCOPE",
        )

        row = {"path": str(relative), "c081_skip_reason": skipped_row["reason"]}
        if path.is_symlink() or not path.is_file():
            row["status"] = "FILE_MISSING_OR_UNSUITABLE"
        else:
            found = set()
            has_variant = False
            digest = hashlib.sha256()
            tail = b""
            bytes_read = 0
            try:
                with path.open("rb") as handle:
                    before = os.fstat(handle.fileno())
                    while chunk := handle.read(4 * 1024 * 1024):
                        digest.update(chunk)
                        bytes_read += len(chunk)
                        window = tail + chunk
                        has_variant = has_variant or variant_token in window
                        for token, candidate in tokens.items():
                            if candidate not in found and token in window:
                                found.add(candidate)
                        tail = window[-overlap:]
                    after = os.fstat(handle.fileno())

                stable = (
                    bytes_read == before.st_size == after.st_size
                    and before.st_mtime_ns == after.st_mtime_ns
                )
                if stable:
                    row.update({
                        "status": "SCANNED",
                        "size_bytes": bytes_read,
                        "file_bytes_sha256": digest.hexdigest(),
                        "variant_found_as_text": has_variant,
                        "candidate_ids_found_as_text": sorted(found),
                        "runtime_binding_verified": False,
                    })
                else:
                    row["status"] = "FILE_CHANGED_DURING_READ"
            except OSError as error:
                row.update({"status": "READ_ERROR", "error_type": type(error).__name__})

        files.append(row)
        print(f"C082 {index}/{len(skipped)} | {row['status']}", flush=True)

    leads = [
        row for row in files
        if row["status"] == "SCANNED"
        and (row["variant_found_as_text"] or row["candidate_ids_found_as_text"])
    ]
    unread = [row for row in files if row["status"] != "SCANNED"]
    result = {
        "cell": "C082",
        "roadmap_action": "R11",
        "roadmap_subtask": "R11.3",
        "status": "C081_SKIPPED_FILES_CHECKED",
        "input_file_hashes": {"C080": c080_sha, "C081": c081_sha},
        "target_file_count": len(skipped),
        "scanned_file_count": len(files) - len(unread),
        "files_still_unread_count": len(unread),
        "textual_lead_file_count": len(leads),
        "files_with_both_variant_and_candidate_mentions": sum(
            bool(row["variant_found_as_text"] and row["candidate_ids_found_as_text"])
            for row in leads
        ),
        "file_results": files,
        "verified_product_runtime_receipts_in_this_cell": 0,
        "new_api_calls": 0,
        "product_regression_tests_executed": 0,
        "selected_version_gate": "BLOCKED",
    }

    data = json.dumps(
        result, ensure_ascii=False, sort_keys=True, separators=(",", ":"),
        allow_nan=False,
    ).encode("utf-8")
    digest = sha(data)
    output = folder / f"C082.R11_skipped_runtime_receipt_files.{digest}.json"
    if not output.exists():
        temporary = None
        try:
            with tempfile.NamedTemporaryFile(dir=folder, prefix="C082.", delete=False) as handle:
                temporary = Path(handle.name)
                handle.write(data)
                handle.flush()
                os.fsync(handle.fileno())
            os.replace(temporary, output)
        finally:
            if temporary is not None:
                temporary.unlink(missing_ok=True)

    require(output.read_bytes() == data, "C082_OUTPUT_RELOAD_MISMATCH")
    print(json.dumps({
        **{key: value for key, value in result.items() if key != "file_results"},
        "lead_paths": [row["path"] for row in leads],
        "unread_files": unread,
        "output": {
            "path": str(output.relative_to(root)),
            "file_bytes_sha256": digest,
            "reload_validation": "PASS",
        },
    }, ensure_ascii=False, indent=2))
    return result


V3_C082 = run_v3_c082()
````

### C085 — ردیابی مسیر تولید پاسخ و ثبت رسید در R11.3

Gate نسخهٔ منتخب همچنان BLOCKED است.

````python
# C085 | R11.3
# C001 provides Path, os, json, hashlib, subprocess, tempfile.

def run_v3_c085():
    def sha(data):
        return hashlib.sha256(data).hexdigest()

    def require(condition, code):
        if not condition:
            raise ValueError(code)

    root = Path(os.environ.get(
        "RAG_FINANCE_ROOT",
        "/Users/macbookpro/AMIR_DATA/00_WORKS/01_PROJECTS/"
        "RAG-projects/RAG_finance",
    )).expanduser().resolve()
    folder = root / "reports/phase2/quality_improvement/NB22E.lab.v3"
    input_sha = "c04219e09e253c1e38412896703611c20f5598d192c443032dddbcca0599e5aa"
    input_path = folder / f"C084.R11_char_rrf_notebook_code.{input_sha}.json"

    raw = input_path.read_bytes()
    require(sha(raw) == input_sha, "C085_C084_HASH_MISMATCH")
    previous = json.loads(raw)
    commit = previous["local_commit"]
    require(
        previous["cell"] == "C084"
        and previous["roadmap_subtask"] == "R11.3"
        and previous["matching_code_cell_count"] == 4
        and len(commit) == 40
        and all(char in "0123456789abcdef" for char in commit),
        "C085_C084_BINDING_MISMATCH",
    )

    def git(*args):
        return subprocess.run(
            ["git", "-C", str(root), *args],
            capture_output=True,
            timeout=120,
        )

    top = git("rev-parse", "--show-toplevel")
    require(
        top.returncode == 0
        and Path(os.fsdecode(top.stdout).strip()).resolve() == root,
        "C085_GIT_ROOT_MISMATCH",
    )

    generator_path = "src/generation/arvan_generator.py"
    generator = git("show", f"{commit}:{generator_path}")

    needles = ["arvan_generator"]
    if generator.returncode == 0:
        # Find top-level class and function names for the caller search.
        # Read source text only. The module is never imported.
        for line in generator.stdout.decode("utf-8").splitlines():
            if line.startswith(("class ", "def ", "async def ")):
                name = line.split("def ", 1)[-1].removeprefix("class ")
                name = name.split("(", 1)[0].split(":", 1)[0].strip()
                if name.isidentifier() and not name.startswith("_"):
                    needles.append(name)
    needles = list(dict.fromkeys(needles))

    grep_args = [part for needle in needles for part in ("-e", needle)]
    found = git(
        "grep", "-I", "-z", "-l", "-F", *grep_args, commit, "--",
        ":(glob)**/*.py", ":(glob)**/*.ipynb",
    )
    require(found.returncode in (0, 1), "C085_GIT_GREP_FAILED")

    prefix = f"{commit}:"
    paths = set()
    for entry in found.stdout.split(b"\0"):
        if entry:
            reference = os.fsdecode(entry)
            require(reference.startswith(prefix), "C085_UNEXPECTED_GREP_FORMAT")
            paths.add(reference[len(prefix):])
    if generator.returncode == 0:
        paths.add(generator_path)

    context_terms = (
        "char_rrf", "receipt", "manifest", "candidate_id", "evidence_id",
        "question", "sources", "response", "sha256", "write_text", "json.dump",
    )
    files = []
    saved_output_only_files = []

    for name in sorted(paths):
        blob = git("show", f"{commit}:{name}")
        require(blob.returncode == 0, f"C085_GIT_SHOW_FAILED: {name}")

        if name.endswith(".ipynb"):
            notebook = json.loads(blob.stdout)
            units = []
            for index, cell in enumerate(notebook["cells"]):
                if cell["cell_type"] != "code":
                    continue
                source = cell.get("source", "")
                units.append((index, source if isinstance(source, str) else "".join(source)))
        else:
            units = [(None, blob.stdout.decode("utf-8"))]

        excerpts = []
        for cell_index, source in units:
            lines = source.splitlines()
            hits = [
                index for index, line in enumerate(lines)
                if any(needle in line for needle in needles)
            ]
            if not hits and name != generator_path:
                continue

            selected = {
                nearby
                for index in hits
                for nearby in range(max(0, index - 3), min(len(lines), index + 4))
            }
            selected.update(
                index for index, line in enumerate(lines)
                if any(term in line.lower() for term in context_terms)
            )
            excerpts.append({
                "cell_index_zero_based": cell_index,
                "source_sha256": sha(source.encode("utf-8")),
                "generator_reference_lines": [index + 1 for index in hits],
                "selected_source_lines": [
                    {"line": index + 1, "text": lines[index]}
                    for index in sorted(selected)
                ],
            })

        if excerpts:
            files.append({
                "path": name,
                "git_blob_bytes_sha256": sha(blob.stdout),
                "code_units": excerpts,
            })
        else:
            # git grep can also hit notebook Markdown, metadata or saved output.
            saved_output_only_files.append(name)

    result = {
        "cell": "C085",
        "roadmap_action": "R11",
        "roadmap_subtask": "R11.3",
        "status": "GENERATOR_SOURCE_AND_CALLER_LEADS_EXTRACTED",
        "input_file_hashes": {"C084": input_sha},
        "local_commit": commit,
        "generator_path_found_at_commit": generator.returncode == 0,
        "search_identifiers": needles,
        "files_with_code_leads": files,
        "files_with_match_outside_inspected_code": saved_output_only_files,
        "runtime_entrypoint_verified": False,
        "product_regression_tests_executed": 0,
        "new_api_calls": 0,
        "selected_version_gate": "BLOCKED",
    }

    data = json.dumps(
        result, ensure_ascii=False, sort_keys=True, separators=(",", ":"),
        allow_nan=False,
    ).encode("utf-8")
    digest = sha(data)
    output = folder / f"C085.R11_generator_and_caller_leads.{digest}.json"

    if not output.exists():
        temporary = None
        try:
            with tempfile.NamedTemporaryFile(
                dir=folder, prefix="C085.", delete=False,
            ) as handle:
                temporary = Path(handle.name)
                handle.write(data)
                handle.flush()
                os.fsync(handle.fileno())
            os.replace(temporary, output)
        finally:
            if temporary is not None:
                temporary.unlink(missing_ok=True)

    require(output.read_bytes() == data, "C085_OUTPUT_RELOAD_MISMATCH")
    print(json.dumps({
        **{key: value for key, value in result.items() if key != "files_with_code_leads"},
        "code_lead_summary": [
            {
                "path": item["path"],
                "matching_code_unit_count": len(item["code_units"]),
            }
            for item in files
        ],
        "output": {
            "path": str(output.relative_to(root)),
            "file_bytes_sha256": digest,
            "reload_validation": "PASS",
        },
    }, ensure_ascii=False, indent=2))
    return result


V3_C085 = run_v3_c085()
````

````python
# C090 | R11.3 | C001 provides Path, os, json, hashlib, re, tempfile.

def run_v3_c090():
    def require(condition, code):
        if not condition:
            raise ValueError(code)

    def sha(data):
        return hashlib.sha256(data).hexdigest()

    def encode(value):
        return json.dumps(
            value, ensure_ascii=False, sort_keys=True,
            separators=(",", ":"), allow_nan=False,
        ).encode("utf-8")

    root = Path(os.environ.get(
        "RAG_FINANCE_ROOT",
        "/Users/macbookpro/AMIR_DATA/00_WORKS/01_PROJECTS/"
        "RAG-projects/RAG_finance",
    )).expanduser().resolve()
    folder = root / "reports/phase2/quality_improvement/NB22E.lab.v3"

    pins = {
        "C036": (
            "C036.R11_expanded_pool_diagnostic",
            "69a6ffdbd965ce8a2a6176ba201445798d1604c84e73954a6a407b7411f2b7b2",
        ),
        "C043": (
            "C043.R11_scenario_review_inputs",
            "811af55fabd63d695bd2767c762055c4997d81e4acd7c36f7bdde717421e6145",
        ),
        "C080": (
            "C080.R11_diagnostic_residuals_and_execution_requirements",
            "15dba7dddba32ec40bd63547637735794932a72c02a970df4e45b1f49678857b",
        ),
    }
    documents = {}
    for cell, (stem, digest) in pins.items():
        raw = (folder / f"{stem}.{digest}.json").read_bytes()
        require(sha(raw) == digest, f"C090_INPUT_HASH_MISMATCH: {cell}")
        documents[cell] = json.loads(raw)

    source = documents["C043"]
    requirements = documents["C080"]
    diagnostic = documents["C036"]
    scope = "FROZEN_CHAR_FUSION_ON_HISTORICAL_POOL50_INCUMBENT"
    require(
        source["scope"] == requirements["scope"] == scope
        and requirements["input_file_hashes"]["C043"] == pins["C043"][1]
        and diagnostic["status"] == "R11_EXPANDED_POOL_DIAGNOSTIC_EXECUTED"
        and all(
            item["selected_version_gate"] == "BLOCKED"
            for item in documents.values()
        ),
        "C090_SCOPE_OR_INPUT_BINDING_MISMATCH",
    )

    response_rows = source["responses"]
    responses = {row["candidate_id"]: row for row in response_rows}
    rank_rows = diagnostic["per_query"]
    rankings = {row["candidate_id"]: row for row in rank_rows}
    scenario_rows = source["scenario_definitions"]["selected_version_execution_requests"]
    scenarios = {row["regression_spec_id"]: row for row in scenario_rows}
    execution_rows = requirements["product_execution_requirements"]
    require(
        len(response_rows) == len(responses) == 34
        and len(rank_rows) == len(rankings) == diagnostic["questions_evaluated"] == 226
        and len(scenario_rows) == len(scenarios) == len(execution_rows) == 36
        and {row["regression_spec_id"] for row in execution_rows} == set(scenarios),
        "C090_POPULATION_MISMATCH",
    )

    by_candidate = {}
    kind_counts = {"ANSWER_DECISION": 0, "CITATION_WITNESS": 0}
    for row in execution_rows:
        scenario_id = row["regression_spec_id"]
        candidate_id = row["candidate_id"]
        kind = row["kind"]
        original = scenarios[scenario_id]
        require(
            candidate_id == original["candidate_id"]
            and kind == original["kind"] and kind in kind_counts
            and row["required_execution_evidence"] == original["required_execution_evidence"]
            and row["product_regression_status"] == "BLOCKED",
            f"C090_SCENARIO_BINDING_MISMATCH: {scenario_id}",
        )
        by_candidate.setdefault(candidate_id, []).append({
            "regression_spec_id": scenario_id,
            "kind": kind,
            "required_execution_evidence": row["required_execution_evidence"],
        })
        kind_counts[kind] += 1

    require(
        set(by_candidate) == set(responses)
        and kind_counts == {"ANSWER_DECISION": 31, "CITATION_WITNESS": 5},
        "C090_TARGET_COVERAGE_MISMATCH",
    )

    header = re.compile(
        r"(?m)^\[منبع ([0-9]+)\]\nشناسه چانک: ([^\n]+)\n\nمتن:\n"
    )
    receipts = []
    matched_ids = []
    unverified_ids = []

    for candidate_id, response in sorted(responses.items()):
        messages = response["messages_used_by_c042"]
        require(
            messages and messages[-1]["role"] == "user",
            f"C090_USER_MESSAGE_MISSING: {candidate_id}",
        )
        prompt = messages[-1]["content"]
        question_section, separator, sources_text = prompt.partition("\nمنابع:\n")
        require(
            bool(separator) and question_section.startswith("سؤال:\n"),
            f"C090_QUESTION_FORMAT_MISMATCH: {candidate_id}",
        )
        question = question_section.removeprefix("سؤال:\n").strip()
        headers = list(header.finditer(sources_text))
        require(
            bool(question)
            and [int(match.group(1)) for match in headers] == list(range(1, 11)),
            f"C090_SOURCE_SEQUENCE_MISMATCH: {candidate_id}",
        )

        ordered_sources = []
        for index, match in enumerate(headers):
            end = headers[index + 1].start() if index + 1 < len(headers) else len(sources_text)
            # Hash the rendered body exactly as delimited by the next source header.
            # This can include prompt separators and is not a corpus text hash.
            body = sources_text[match.end():end]
            evidence_id = match.group(2).strip()
            require(
                bool(evidence_id) and bool(body.strip()),
                f"C090_EMPTY_SOURCE: {candidate_id}",
            )
            ordered_sources.append({
                "rank": int(match.group(1)),
                "evidence_id": evidence_id,
                "rendered_source_body_sha256": sha(body.encode("utf-8")),
            })

        ordered_ids = [item["evidence_id"] for item in ordered_sources]
        require(
            len(set(ordered_ids)) == 10,
            f"C090_DUPLICATE_SOURCE_ID: {candidate_id}",
        )
        if candidate_id in rankings:
            require(
                ordered_ids == rankings[candidate_id]["expanded_top10"],
                f"C090_C036_ORDER_MISMATCH: {candidate_id}",
            )
            rank_status = "MATCHED_C036_HISTORICAL_POOL50_DIAGNOSTIC_TOP10"
            matched_ids.append(candidate_id)
        else:
            rank_status = "SENT_ORDER_RECORDED_RANK_ORIGIN_UNVERIFIED"
            unverified_ids.append(candidate_id)

        receipts.append({
            "candidate_id": candidate_id,
            "scenarios": by_candidate[candidate_id],
            "scope": scope,
            "receipt_timing": "RECONSTRUCTED_AFTER_EXECUTION_FROM_C043",
            "question_sha256": sha(question.encode("utf-8")),
            "messages_canonical_sha256": sha(encode(messages)),
            "user_message_sha256": sha(prompt.encode("utf-8")),
            "generation_input_sha256_as_recorded": response["generation_input_sha256"],
            "ordered_sources": ordered_sources,
            "rank_provenance_status": rank_status,
            "response_model_as_recorded": response["response_model"],
            "response_bytes_sha256_as_recorded": response["response_bytes_sha256"],
            "response_content_sha256": sha(response["response_content"].encode("utf-8")),
            "product_runtime_binding_verified": False,
        })

    require(
        len(receipts) == 34 and len(matched_ids) == 18 and len(unverified_ids) == 16,
        "C090_EXPECTED_18_MATCHED_AND_16_UNVERIFIED",
    )

    result = {
        "cell": "C090",
        "roadmap_action": "R11",
        "roadmap_subtask": "R11.3",
        "status": "FIXED_INPUT_DIAGNOSTIC_RECEIPTS_RECONSTRUCTED",
        "scope": scope,
        "input_file_hashes": {cell: pin[1] for cell, pin in pins.items()},
        "candidate_receipt_count": len(receipts),
        "scenario_counts": kind_counts,
        "c036_order_match_count": len(matched_ids),
        "c036_order_matched_candidate_ids": matched_ids,
        "rank_origin_unverified_count": len(unverified_ids),
        "rank_origin_unverified_candidate_ids": unverified_ids,
        "rank_reconstruction_for_unverified_candidates": "DEFERRED",
        "future_dispatch_requires_pre_generation_receipt": True,
        "comparable_component_baseline_verified_in_this_cell": False,
        "end_to_end_blocker": "PRODUCT_ENTRYPOINT_BINDING_TO_SELECTED_CHAR_RRF_NOT_VERIFIED",
        "new_api_calls": 0,
        "ranking_runs_executed": 0,
        "product_regression_tests_executed": 0,
        "historical_gate": "FAIL",
        "selected_version_gate": "BLOCKED",
        "receipts": receipts,
    }

    data = encode(result)
    digest = sha(data)
    output = folder / f"C090.R11_fixed_input_diagnostic_receipts.{digest}.json"
    if not output.exists():
        temporary = None
        try:
            with tempfile.NamedTemporaryFile(
                dir=folder, prefix="C090.", delete=False,
            ) as handle:
                temporary = Path(handle.name)
                handle.write(data)
                handle.flush()
                os.fsync(handle.fileno())
            os.replace(temporary, output)
        finally:
            if temporary is not None:
                temporary.unlink(missing_ok=True)

    require(output.read_bytes() == data, "C090_OUTPUT_RELOAD_MISMATCH")
    print(json.dumps({
        **{key: value for key, value in result.items() if key != "receipts"},
        "output": {
            "path": str(output.relative_to(root)),
            "file_bytes_sha256": digest,
            "reload_validation": "PASS",
        },
    }, ensure_ascii=False, indent=2))
    return result


V3_C090 = run_v3_c090()
````

### C092 — بررسی صلاحیت شواهد تاریخی به‌عنوان خط پایهٔ مؤلفه

````python
# C094 | R11.3 | C001 provides Path, os, json, hashlib, tempfile.

def run_v3_c094():
    def sha(data):
        return hashlib.sha256(data).hexdigest()

    def require(condition, code):
        if not condition:
            raise ValueError(code)

    root = Path(os.environ.get(
        "RAG_FINANCE_ROOT",
        "/Users/macbookpro/AMIR_DATA/00_WORKS/01_PROJECTS/"
        "RAG-projects/RAG_finance",
    )).expanduser().resolve()
    folder = root / "reports/phase2/quality_improvement/NB22E.lab.v3"

    c093_sha = "9e0d4755409a97f0f2b64f40d8e44d3e947b3d9f6afd4f58bb1dd81b69dfaf53"
    c069_sha = "4dae1720b3c5d18e07eb89d61722ad2e2ee9b93ec2b314eb48a07b4a778b0ae5"
    c081_sha = "0c776c3b64dc517e752f9c22ff7eb5a9456d34290c00f3c9713644b5cf9db73d"

    prior_path = folder / f"C093.R11_component_baseline_receipt_check.{c093_sha}.json"
    prior_bytes = prior_path.read_bytes()
    require(sha(prior_bytes) == c093_sha, "C094_C093_HASH_MISMATCH")
    prior = json.loads(prior_bytes)
    require(
        prior["cell"] == "C093"
        and prior["roadmap_subtask"] == "R11.3"
        and prior["scenario_count"] == 36
        and prior["component_comparison_status"] == "NOT_COMPARABLE_NOT_PASS"
        and prior["next_component_dependency"]
        == "OBTAIN_VERIFIABLE_BASELINE_REPORT_AND_ROW_LEVEL_RECEIPTS",
        "C094_C093_STATE_MISMATCH",
    )

    pins = {
        "NB21C_C069_METRIC_TABLE": (
            "reports/phase2/evaluation/NB21C.C069.action39_offline_final/"
            f"C069.metric_table.{c069_sha}.json",
            c069_sha,
        ),
        "NB21C_C081_GATE_REPORT": (
            "reports/phase2/evaluation/NB21C.C081.action45_regression_gate/"
            f"C081.report.{c081_sha}.json",
            c081_sha,
        ),
    }

    def iter_receipts(document):
        stack = [("", document)]
        while stack:
            pointer, value = stack.pop()
            if isinstance(value, dict):
                path = value.get("path")
                digest = value.get("file_bytes_sha256")
                if (
                    isinstance(path, str) and path
                    and isinstance(digest, str) and len(digest) == 64
                    and all(char in "0123456789abcdef" for char in digest)
                ):
                    yield {
                        "json_pointer": pointer,
                        "path": path,
                        "file_bytes_sha256": digest,
                    }
                for key, child in value.items():
                    escaped = key.replace("~", "~0").replace("/", "~1")
                    stack.append((f"{pointer}/{escaped}", child))
            elif isinstance(value, list):
                stack.extend(
                    (f"{pointer}/{index}", child)
                    for index, child in enumerate(value)
                )

    terms = (
        "baseline", "accepted", "acceptance", "row_level", "generation",
        "answer", "citation", "grounded", "attribution", "provenance",
    )
    inspected = {}
    leads = []

    for label, (relative_path, expected_sha) in pins.items():
        check = {
            "path": relative_path,
            "expected_file_bytes_sha256": expected_sha,
        }
        inspected[label] = check
        path = root / relative_path
        if not path.is_file():
            check["status"] = "FILE_UNAVAILABLE"
            continue

        raw = path.read_bytes()
        actual_sha = sha(raw)
        check["actual_file_bytes_sha256"] = actual_sha
        if actual_sha != expected_sha:
            check["status"] = "FILE_HASH_MISMATCH"
            continue

        document = json.loads(raw)
        receipts = sorted(
            iter_receipts(document),
            key=lambda row: (row["json_pointer"], row["path"]),
        )
        check["status"] = "HASH_VERIFIED_AND_INSPECTED"
        check["all_embedded_receipts"] = receipts
        for receipt in receipts:
            searchable = f"{receipt['json_pointer']} {receipt['path']}".lower()
            matched_terms = [term for term in terms if term in searchable]
            if matched_terms:
                leads.append({
                    "source_document": label,
                    **receipt,
                    "matched_terms": matched_terms,
                    "baseline_acceptance_verified": False,
                })

    verified_count = sum(
        row["status"] == "HASH_VERIFIED_AND_INSPECTED"
        for row in inspected.values()
    )
    result = {
        "cell": "C094",
        "roadmap_action": "R11",
        "roadmap_subtask": "R11.3",
        "status": "PINNED_HISTORICAL_BASELINE_REFERENCES_CHECKED",
        "input_file_hashes": {"C093": c093_sha},
        "scenario_count": prior["scenario_count"],
        "pinned_documents_verified": verified_count,
        "documents": inspected,
        "targeted_receipt_lead_count": len(leads),
        "targeted_receipt_leads": leads,
        "comparable_component_baselines_verified": 0,
        "component_comparison_status": prior["component_comparison_status"],
        "end_to_end_blocker": prior["end_to_end_blocker"],
        "rank_reconstruction_for_unverified_candidates": "DEFERRED",
        "new_api_calls": 0,
        "ranking_runs_executed": 0,
        "product_regression_tests_executed": 0,
        "historical_gate": "FAIL",
        "selected_version_gate": "BLOCKED",
    }

    data = json.dumps(
        result, ensure_ascii=False, sort_keys=True,
        separators=(",", ":"), allow_nan=False,
    ).encode("utf-8")
    digest = sha(data)
    output = folder / f"C094.R11_targeted_baseline_reference_trace.{digest}.json"

    if not output.exists():
        temporary = None
        try:
            with tempfile.NamedTemporaryFile(
                dir=folder, prefix="C094.", delete=False,
            ) as handle:
                temporary = Path(handle.name)
                handle.write(data)
                handle.flush()
                os.fsync(handle.fileno())
            os.replace(temporary, output)
        finally:
            if temporary is not None:
                temporary.unlink(missing_ok=True)

    require(output.read_bytes() == data, "C094_OUTPUT_RELOAD_MISMATCH")
    print(json.dumps({
        **{key: value for key, value in result.items() if key != "documents"},
        "document_statuses": {
            label: row["status"] for label, row in inspected.items()
        },
        "output": {
            "path": str(output.relative_to(root)),
            "file_bytes_sha256": digest,
            "reload_validation": "PASS",
        },
    }, ensure_ascii=False, indent=2))
    return result


V3_C094 = run_v3_c094()
````

````python
# C102 | R11.3 | Run after C001, C086 and C101.

def run_v3_c102():
    def require(condition, code):
        if not condition:
            raise ValueError(code)

    def sha(data):
        return hashlib.sha256(data).hexdigest()

    def encode(value):
        return json.dumps(
            value, ensure_ascii=False, sort_keys=True,
            separators=(",", ":"), allow_nan=False,
        ).encode("utf-8")

    root = Path(os.environ.get(
        "RAG_FINANCE_ROOT",
        "/Users/macbookpro/AMIR_DATA/00_WORKS/01_PROJECTS/RAG-projects/RAG_finance",
    )).expanduser().resolve()
    folder = root / "reports/phase2/quality_improvement/NB22E.lab.v3"

    pins = {
        "C086": (
            "C086.R11_pipeline_evidence_entrypoint",
            "3102977bcd69042dd07b26763f81b069c72a149e66bd8e17bdffc26abd1bd861",
        ),
        "C101": (
            "C101.R11_frozen_char_rrf_fixture",
            "406a00f8bca950803929b08ef48cc152bd8555927bd9b831fc4483d4233fba7c",
        ),
    }

    documents = {}
    for cell, (stem, digest) in pins.items():
        raw = (folder / f"{stem}.{digest}.json").read_bytes()
        require(sha(raw) == digest, f"C102_INPUT_HASH_MISMATCH: {cell}")
        document = json.loads(raw)
        require(
            document["cell"] == cell and document["roadmap_subtask"] == "R11.3",
            f"C102_INPUT_IDENTITY_MISMATCH: {cell}",
        )
        documents[cell] = document

    pipeline = documents["C086"]
    fixture = documents["C101"]
    ordered_ids = fixture["fused_order"]
    require(
        pipeline["local_commit"] == fixture["local_commit"]
        and fixture["status"] == "FROZEN_CHAR_RRF_SYNTHETIC_FIXTURE_PASSED"
        and fixture["selected_version_gate"] == "BLOCKED"
        and ordered_ids == ["char-b", "base-only", "char-a"],
        "C102_FIXTURE_BINDING_MISMATCH",
    )

    method_names = {"ask", "build_context", "format_sources"}
    methods = [
        item for item in pipeline["pipeline_methods"] if item["name"] in method_names
    ]
    require(
        len(methods) == 3 and {item["name"] for item in methods} == method_names,
        "C102_METHOD_SET_MISMATCH",
    )
    namespace = {}
    for method in methods:
        name = method["name"]
        exec(compile(method["source"], f"<C086:{name}>", "exec"), namespace)

    texts = {
        "char-a": "كتاب",
        "char-b": "کتاب در کنار میز",
        "base-only": "باغ",
    }
    chunks = [
        {
            "chunk_id": chunk_id,
            "title": f"منبع ساختگی {index}",
            "url": f"fixture://{chunk_id}",
            "text": texts[chunk_id],
            "score": None,
        }
        for index, chunk_id in enumerate(ordered_ids, start=1)
    ]
    question = "كِتاب"
    answer = "در منابع موجود پاسخ کافی پیدا نشد."
    runtime_identity = {
        "scope": "OFFLINE_SYNTHETIC_FIXTURE",
        "pipeline_commit": pipeline["local_commit"],
        "rank_order_reference_sha256": pins["C101"][1],
    }

    def write_verified(path, data):
        if not path.exists():
            temporary = None
            try:
                with tempfile.NamedTemporaryFile(
                    dir=path.parent, prefix=f".{path.stem}.", delete=False,
                ) as handle:
                    temporary = Path(handle.name)
                    handle.write(data)
                    handle.flush()
                    os.fsync(handle.fileno())
                os.replace(temporary, path)
            finally:
                if temporary is not None:
                    temporary.unlink(missing_ok=True)
        require(path.read_bytes() == data, f"C102_FILE_RELOAD_MISMATCH: {path.name}")

    # A fresh directory proves that this invocation writes the receipt.
    # Only the verified, content-addressed receipt is retained afterward.
    with tempfile.TemporaryDirectory(dir=folder, prefix=".C102-fixture-") as work:
        staged_path = Path(work) / "input_receipt.json"
        events = []

        class FixtureEmbedder:
            def embed_query(self, received_question):
                require(received_question == question, "C102_EMBED_QUESTION_MISMATCH")
                return [1.0]

        class FixtureRetriever:
            def search(self, *, query_vector, top_k):
                require(
                    query_vector == [1.0] and top_k == len(chunks),
                    "C102_RETRIEVER_ARGUMENT_MISMATCH",
                )
                return [dict(chunk) for chunk in chunks]

        class FixtureGenerator:
            def generate_answer(self, *, question, context):
                require(
                    events == ["INPUT_RECEIPT_WRITTEN"]
                    and staged_path.is_file(),
                    "C102_INPUT_RECEIPT_NOT_PRESENT_BEFORE_GENERATION",
                )
                receipt = json.loads(staged_path.read_bytes())
                rendered_ids = [
                    line.removeprefix("شناسه چانک: ")
                    for line in context.splitlines()
                    if line.startswith("شناسه چانک: ")
                ]
                require(
                    receipt["question_sha256"] == sha(question.encode("utf-8"))
                    and receipt["ordered_chunk_ids"] == rendered_ids == ordered_ids
                    and receipt["rendered_context_sha256"] == sha(context.encode("utf-8"))
                    and receipt["ordered_chunk_text_sha256"] == [
                        sha(texts[chunk_id].encode("utf-8"))
                        for chunk_id in ordered_ids
                    ],
                    "C102_GENERATOR_INPUT_RECEIPT_MISMATCH",
                )
                events.append("GENERATOR_VERIFIED_INPUT_RECEIPT")
                return answer

        class FixturePipeline:
            ask = namespace["ask"]
            format_sources = namespace["format_sources"]
            use_cache = False
            cache = None
            embedding_model = "offline-fixture"
            generation_model = "offline-fixture"
            embedder = FixtureEmbedder()
            retriever = FixtureRetriever()
            generator = FixtureGenerator()

            def build_context(self, retrieved_chunks):
                context = namespace["build_context"](self, retrieved_chunks)
                receipt = {
                    "candidate_id": "fixture-c102",
                    "question_sha256": sha(question.encode("utf-8")),
                    "selected_variant": "char_rrf",
                    "runtime_identity": runtime_identity,
                    "ordered_chunk_ids": [item["chunk_id"] for item in retrieved_chunks],
                    "ordered_chunk_text_sha256": [
                        sha(item["text"].encode("utf-8")) for item in retrieved_chunks
                    ],
                    "rendered_context_sha256": sha(context.encode("utf-8")),
                }
                write_verified(staged_path, encode(receipt))
                events.append("INPUT_RECEIPT_WRITTEN")
                return context

        output = FixturePipeline().ask(question, top_k=len(chunks))
        require(
            events == ["INPUT_RECEIPT_WRITTEN", "GENERATOR_VERIFIED_INPUT_RECEIPT"]
            and output["answer"] == answer
            and output["from_cache"] is False
            and output["sources"] == output["retrieved_chunks"] == [],
            "C102_ABSTENTION_OUTPUT_MISMATCH",
        )
        receipt_bytes = staged_path.read_bytes()

    receipt_sha = sha(receipt_bytes)
    receipt_path = folder / f"C102.R11_fixture_input_receipt.{receipt_sha}.json"
    write_verified(receipt_path, receipt_bytes)

    result = {
        "cell": "C102",
        "roadmap_action": "R11",
        "roadmap_subtask": "R11.3",
        "status": "OFFLINE_PRE_GENERATION_RECEIPT_AND_ABSTENTION_FIXTURE_PASSED",
        "scope": "REVIEWED_ASK_WITH_SYNTHETIC_RETRIEVER_AND_GENERATOR",
        "input_file_hashes": {cell: pin[1] for cell, pin in pins.items()},
        "ordered_chunk_ids_passed_to_generator": ordered_ids,
        "events": events,
        "input_receipt": {
            "path": str(receipt_path.relative_to(root)),
            "file_bytes_sha256": receipt_sha,
        },
        "response_binding": {
            "input_receipt_file_bytes_sha256": receipt_sha,
            "answer_text_utf8_sha256": sha(output["answer"].encode("utf-8")),
        },
        "rank_reconstruction_for_unverified_candidates": "DEFERRED",
        "product_runtime_binding_verified": False,
        "end_to_end_blocker": fixture["end_to_end_blocker"],
        "new_api_calls": 0,
        "product_regression_tests_executed": 0,
        "historical_gate": fixture["historical_gate"],
        "selected_version_gate": "BLOCKED",
    }

    data = encode(result)
    digest = sha(data)
    path = folder / f"C102.R11_pre_generation_receipt_fixture.{digest}.json"
    write_verified(path, data)
    print(json.dumps({
        **result,
        "output": {
            "path": str(path.relative_to(root)),
            "file_bytes_sha256": digest,
            "reload_validation": "PASS",
        },
    }, ensure_ascii=False, indent=2))
    return result


V3_C102 = run_v3_c102()
````

### C106 — تعیین شاخهٔ واقعاً ذخیره‌شدهٔ رتبهٔ پایه در R11.3

احراز نشده است. سل آفلاین است و Gate نسخهٔ منتخب BLOCKED است.

</div>

````python
# C106 | R11.3 | Run after C001.

def run_v3_c106():
    def require(condition, code):
        if not condition:
            raise ValueError(code)

    def sha(data):
        return hashlib.sha256(data).hexdigest()

    root = Path(os.environ.get(
        "RAG_FINANCE_ROOT",
        "/Users/macbookpro/AMIR_DATA/00_WORKS/01_PROJECTS/RAG-projects/RAG_finance",
    )).expanduser().resolve()
    folder = root / "reports/phase2/quality_improvement/NB22E.lab.v3"
    nb22c = root / "reports/phase2/quality_improvement/NB22C.v1"

    c104_sha = "a89271c9b6cf5cd6ec5cf90c4365600a42434264c151179a18b7732b71813774"
    ranking_sha = "5bc216aae839bc80cc2f184ca3a3cd3e4ac33657655c3bffb4fac410e96e9190"
    final_sha = "1797d76fe08136a50ffa846da39858f29838a9bc8a3dc3d1cad120789ccc53ea"

    def read_pinned(path, digest):
        raw = path.read_bytes()
        require(sha(raw) == digest, f"C106_INPUT_HASH_MISMATCH: {path.name}")
        return json.loads(raw)

    prior = read_pinned(
        folder / f"C104.R11_nb22c_incumbent_producer_trace.{c104_sha}.json",
        c104_sha,
    )
    ranking_path = nb22c / f"A05.retained_development_rankings.{ranking_sha}.jsonl"
    final_path = nb22c / f"A06.final_report.{final_sha}.json"
    require(
        prior["cell"] == "C104"
        and prior["roadmap_subtask"] == "R11.3"
        and prior["selected_version_gate"] == "BLOCKED"
        and prior["incumbent_artifact"]["path_as_recorded"]
        == str(ranking_path.relative_to(root))
        and prior["incumbent_artifact"]["file_bytes_sha256_as_recorded"] == ranking_sha,
        "C106_C104_BINDING_MISMATCH",
    )
    report = read_pinned(final_path, final_sha)

    file_hash = hashlib.sha256()
    variant_counts = {}
    candidate_ids = set()
    with ranking_path.open("rb") as handle:
        for line_number, raw in enumerate(handle, start=1):
            file_hash.update(raw)
            if not raw.strip():
                continue
            row = json.loads(raw)
            candidate_id = row["candidate_id"]
            variant = row["variant"]
            order = row["ranked_chunk_ids"]
            require(
                row["split"] == "development"
                and isinstance(candidate_id, str) and bool(candidate_id)
                and candidate_id not in candidate_ids
                and isinstance(variant, str) and bool(variant)
                and isinstance(order, list) and bool(order)
                and all(isinstance(chunk_id, str) and chunk_id for chunk_id in order)
                and len(order) == len(set(order)),
                f"C106_INVALID_RANKING_ROW: {line_number}",
            )
            candidate_ids.add(candidate_id)
            variant_counts[variant] = variant_counts.get(variant, 0) + 1

    require(file_hash.hexdigest() == ranking_sha, "C106_A05_HASH_MISMATCH")
    require(len(variant_counts) == 1, "C106_EXPECTED_ONE_RETAINED_VARIANT")
    retained = next(iter(variant_counts))
    fallback = "NB22B_SELECTED_cached_rrf_k10_w0.75"

    # Find objects that carry both the A05 receipt and the decision fields.
    stack = [("", report)]
    decisions = []
    expected_path = str(ranking_path.relative_to(root))
    while stack:
        pointer, item = stack.pop()
        if isinstance(item, dict):
            receipt = item.get("retained_rankings")
            if (
                isinstance(receipt, dict)
                and receipt.get("path") == expected_path
                and receipt.get("file_bytes_sha256") == ranking_sha
                and "retained_variant" in item
                and "candidate_passes_internal_guard" in item
            ):
                decisions.append({
                    "json_pointer": pointer,
                    "retained_variant": item["retained_variant"],
                    "candidate_passes_internal_guard": item[
                        "candidate_passes_internal_guard"
                    ],
                    "frozen_tuning_candidate": item.get("frozen_tuning_candidate"),
                })
            for key, value in item.items():
                escaped = key.replace("~", "~0").replace("/", "~1")
                stack.append((f"{pointer}/{escaped}", value))
        elif isinstance(item, list):
            stack.extend(
                (f"{pointer}/{index}", value)
                for index, value in enumerate(item)
            )

    expected_pass = retained != fallback
    for decision in decisions:
        require(
            decision["retained_variant"] == retained
            and decision["candidate_passes_internal_guard"] is expected_pass
            and (
                not expected_pass
                or decision["frozen_tuning_candidate"] == retained
            ),
            "C106_A06_DECISION_CONFLICT_WITH_A05",
        )

    result = {
        "cell": "C106",
        "roadmap_action": "R11",
        "roadmap_subtask": "R11.3",
        "status": (
            "A05_RETAINED_BRANCH_VERIFIED_AGAINST_A06"
            if decisions else "A05_RETAINED_LABEL_VERIFIED_A06_DECISION_LINK_UNRESOLVED"
        ),
        "input_file_hashes": {"C104": c104_sha, "NB22C_A05": ranking_sha, "NB22C_A06": final_sha},
        "development_candidate_count": len(candidate_ids),
        "variant_counts": variant_counts,
        "retained_variant": retained,
        "branch_from_a05_label": (
            "NB22B_INCUMBENT_RETAINED" if retained == fallback
            else "NB22C_CANDIDATE_RETAINED"
        ),
        "a06_decision_links": decisions,
        "incumbent_order_equivalence_verified": False,
        "product_runtime_binding_verified": False,
        "end_to_end_blocker": prior["end_to_end_blocker"],
        "rank_reconstruction_for_unverified_candidates": "DEFERRED",
        "new_api_calls": 0,
        "ranking_runs_executed": 0,
        "product_regression_tests_executed": 0,
        "historical_gate": prior["historical_gate"],
        "selected_version_gate": prior["selected_version_gate"],
    }

    data = json.dumps(
        result, ensure_ascii=False, sort_keys=True, separators=(",", ":"),
        allow_nan=False,
    ).encode("utf-8")
    digest = sha(data)
    output = folder / f"C106.R11_nb22c_retained_branch.{digest}.json"

    if not output.exists():
        temporary = None
        try:
            with tempfile.NamedTemporaryFile(
                dir=folder, prefix="C106.", delete=False,
            ) as handle:
                temporary = Path(handle.name)
                handle.write(data)
                handle.flush()
                os.fsync(handle.fileno())
            os.replace(temporary, output)
        finally:
            if temporary is not None:
                temporary.unlink(missing_ok=True)

    require(output.read_bytes() == data, "C106_OUTPUT_RELOAD_MISMATCH")
    print(json.dumps({
        **result,
        "output": {
            "path": str(output.relative_to(root)),
            "file_bytes_sha256": digest,
            "reload_validation": "PASS",
        },
    }, ensure_ascii=False, indent=2))
    return result


V3_C106 = run_v3_c106()
````

````python
# C109 | R11.3 | Run after C001. No new imports.

def run_v3_c109():
    def require(condition, code):
        if not condition:
            raise ValueError(code)

    def sha(raw):
        return hashlib.sha256(raw).hexdigest()

    def encode(value):
        return json.dumps(
            value, ensure_ascii=False, sort_keys=True,
            separators=(",", ":"), allow_nan=False,
        ).encode("utf-8")

    root = Path(os.environ.get(
        "RAG_FINANCE_ROOT",
        "/Users/macbookpro/AMIR_DATA/00_WORKS/01_PROJECTS/RAG-projects/RAG_finance",
    )).expanduser().resolve()
    folder = root / "reports/phase2/quality_improvement/NB22E.lab.v3"
    pins = {
        "C086": (
            "C086.R11_pipeline_evidence_entrypoint",
            "3102977bcd69042dd07b26763f81b069c72a149e66bd8e17bdffc26abd1bd861",
        ),
        "C100": (
            "C100.R11_frozen_ranker_source",
            "df56d2fc10c834c56bdc957aa90bfef3c731fbb3bdb935617604854864b1f68b",
        ),
        "C107": (
            "C107.R11_nb22c_quota_incumbent_producer",
            "227e24d637bba5c47900296577ca494e4f6736789b463d46dbc951b705066279",
        ),
        "C108": (
            "C108.R11_nb22c_quota100_fixture",
            "b6a5539ebd80f2107015d4dc1c7bbd96653ffc1a91d63aff1e0dc997a25cc74c",
        ),
    }
    documents = {}
    for cell, (stem, digest) in pins.items():
        raw = (folder / f"{stem}.{digest}.json").read_bytes()
        require(sha(raw) == digest, f"C109_INPUT_HASH_MISMATCH: {cell}")
        document = json.loads(raw)
        require(
            document["cell"] == cell
            and document["roadmap_subtask"] == "R11.3"
            and document["selected_version_gate"] == "BLOCKED",
            f"C109_INPUT_IDENTITY_MISMATCH: {cell}",
        )
        documents[cell] = document

    pipeline = documents["C086"]
    ranker = documents["C100"]
    producer = documents["C107"]
    prior = documents["C108"]
    commit = producer["local_commit"]
    require(
        all(item["local_commit"] == commit for item in documents.values())
        and prior["input_file_hashes"]["C107"] == pins["C107"][1]
        and prior["status"] == "PINNED_QUOTA100_SYNTHETIC_FIXTURE_PASSED",
        "C109_PRIOR_BINDING_MISMATCH",
    )

    def pinned_source(document, index):
        matches = [
            row for row in document["selected_code_cells"]
            if row["cell_index_zero_based"] == index
        ]
        require(len(matches) == 1, f"C109_SOURCE_CELL_NOT_UNIQUE: {index}")
        item = matches[0]
        source = item["source"]
        require(
            sha(source.encode("utf-8")) == item["source_sha256"],
            f"C109_SOURCE_HASH_MISMATCH: {index}",
        )
        return source

    def load_function(source, name, namespace, expected_sha=None):
        nodes = [
            node for node in ast.parse(source).body
            if isinstance(node, ast.FunctionDef) and node.name == name
        ]
        require(len(nodes) == 1, f"C109_FUNCTION_NOT_UNIQUE: {name}")
        node = nodes[0]
        text = ast.get_source_segment(source, node)
        if expected_sha is not None:
            require(
                sha(text.encode("utf-8")) == expected_sha,
                f"C109_FUNCTION_HASH_MISMATCH: {name}",
            )
        module = ast.Module(body=[node], type_ignores=[])
        exec(compile(module, f"<C109:{name}>", "exec"), namespace)
        return namespace[name]

    # Inspect the pinned notebook source without running its cells.
    notebook_raw = subprocess.check_output(
        ["git", "-C", str(root), "show", f"{commit}:{producer['notebook_path']}"],
        timeout=60,
    )
    require(
        sha(notebook_raw) == producer["notebook_bytes_sha256"],
        "C109_NB22C_NOTEBOOK_HASH_MISMATCH",
    )
    upstream_leads = []
    markers = (
        "hybrid100", "local100", "hybrid50", "incumbent",
        "rank_ids(", "score_cache",
    )
    for index, cell in enumerate(json.loads(notebook_raw)["cells"]):
        if cell["cell_type"] != "code" or index >= 14:
            continue
        raw_source = cell.get("source", "")
        source = raw_source if isinstance(raw_source, str) else "".join(raw_source)
        lines = [
            {"line": number, "text": line}
            for number, line in enumerate(source.splitlines(), 1)
            if any(marker in line.lower() for marker in markers)
        ]
        if lines:
            upstream_leads.append({
                "cell_index_zero_based": index,
                "source_sha256": sha(source.encode("utf-8")),
                "lines": lines,
            })

    quota = load_function(
        pinned_source(producer, 6),
        "derived_order",
        {"require": require},
        prior["derived_order_source_sha256"],
    )
    spec = prior["retained_variant"]
    require(
        spec == {
            "name": "quota100_k10_w0.75_n1",
            "family": "quota100",
            "k": 10,
            "weight": 0.75,
            "newcomer_limit": 1,
        },
        "C109_QUOTA_SPEC_MISMATCH",
    )

    config_nodes = [
        node.value for node in ast.parse(pinned_source(ranker, 4)).body
        if isinstance(node, ast.Assign)
        and any(isinstance(target, ast.Name) and target.id == "CONFIG"
                for target in node.targets)
    ]
    require(len(config_nodes) == 1, "C109_EXPECTED_ONE_FROZEN_CONFIG")
    config = ast.literal_eval(config_nodes[0])
    char_specs = [item for item in config["ranking_specs"] if item["name"] == "char_rrf"]
    require(
        char_specs == [{"name": "char_rrf", "channels": ["char"]}]
        and config["rrf_k"] == 10
        and config["incumbent_rank_weight"] == 0.75,
        "C109_CHAR_SPEC_MISMATCH",
    )
    fuse = load_function(
        pinned_source(ranker, 12), "lexical_fusion",
        {"CONFIG": config, "collections": collections},
    )

    hybrid = [f"fixture-{index:03d}" for index in range(100)]
    local = [hybrid[1], hybrid[0], *hybrid[2:]]
    incumbent = hybrid[50:]
    ranked = quota(hybrid, local, incumbent, spec)
    reverse_weight = quota(hybrid, local, incumbent, {**spec, "weight": 0.25})
    expected_quota_top10 = [hybrid[1], *hybrid[50:59]]
    require(
        ranked[:10] == expected_quota_top10
        and reverse_weight[:10] == [hybrid[0], *hybrid[50:59]]
        and len(ranked) == len(set(ranked)) == 100
        and set(ranked) == set(hybrid)
        and len(set(ranked[:10]) - set(incumbent)) == 1,
        "C109_WEIGHT_OR_QUOTA_MISMATCH",
    )

    # Char-only evidence is synthetic. The real, pinned fusion runs here.
    char_order = [ranked[1], "fixture-char-only"]
    fused = fuse(ranked, {"char": char_order}, char_specs[0])
    require(
        fused[:3] == [ranked[1], ranked[0], ranked[2]]
        and len(fused) == len(set(fused)) == 101
        and set(fused) == set(ranked) | set(char_order),
        "C109_CHAINED_FUSION_MISMATCH",
    )

    method_names = {"ask", "build_context", "format_sources"}
    methods = {
        item["name"]: item["source"]
        for item in pipeline["pipeline_methods"] if item["name"] in method_names
    }
    require(set(methods) == method_names, "C109_PIPELINE_METHOD_MISSING")
    namespace = {}
    for name, source in methods.items():
        exec(compile(source, f"<C086:{name}>", "exec"), namespace)

    def write_verified(path, data):
        if not path.exists():
            temporary = None
            try:
                with tempfile.NamedTemporaryFile(
                    dir=path.parent, prefix=".C109-", delete=False,
                ) as handle:
                    temporary = Path(handle.name)
                    handle.write(data)
                    handle.flush()
                    os.fsync(handle.fileno())
                try:
                    os.link(temporary, path)
                except FileExistsError:
                    pass
            finally:
                if temporary is not None:
                    temporary.unlink(missing_ok=True)
        require(path.read_bytes() == data, f"C109_RELOAD_MISMATCH: {path.name}")

    question = "سؤال ساختگی C109"
    abstention = "در منابع موجود پاسخ کافی پیدا نشد."
    ordered_ids = fused[:10]
    chunks = [
        {
            "chunk_id": chunk_id,
            "title": f"منبع ساختگی {rank}",
            "url": f"fixture://{chunk_id}",
            "text": f"متن ساختگی {chunk_id}",
            "score": None,
        }
        for rank, chunk_id in enumerate(ordered_ids, 1)
    ]
    expected_text_hashes = [sha(item["text"].encode("utf-8")) for item in chunks]

    events = []
    with tempfile.TemporaryDirectory(dir=folder, prefix=".C109-fixture-") as work:
        receipt_path = Path(work) / "input.json"

        class FixtureEmbedder:
            def embed_query(self, received_question):
                require(received_question == question, "C109_EMBED_QUESTION_MISMATCH")
                return [1.0]

        class FixtureRetriever:
            def search(self, *, query_vector, top_k):
                require(
                    query_vector == [1.0] and top_k == len(chunks),
                    "C109_RETRIEVER_ARGUMENT_MISMATCH",
                )
                return [dict(item) for item in chunks]

        class FixtureGenerator:
            def generate_answer(self, *, question, context):
                require(events == ["INPUT_RECEIPT_WRITTEN"], "C109_RECEIPT_TIMING_MISMATCH")
                receipt = json.loads(receipt_path.read_bytes())
                rendered_ids = [
                    line.removeprefix("شناسه چانک: ")
                    for line in context.splitlines()
                    if line.startswith("شناسه چانک: ")
                ]
                require(
                    receipt["question_sha256"] == sha(question.encode("utf-8"))
                    and receipt["ordered_chunk_ids"] == rendered_ids == ordered_ids
                    and receipt["ordered_chunk_text_sha256"] == expected_text_hashes
                    and receipt["rendered_context_sha256"] == sha(context.encode("utf-8")),
                    "C109_GENERATOR_INPUT_MISMATCH",
                )
                events.append("GENERATOR_VERIFIED_INPUT_RECEIPT")
                return abstention

        class FixturePipeline:
            ask = namespace["ask"]
            format_sources = namespace["format_sources"]
            use_cache = False
            cache = None
            embedding_model = "offline-fixture"
            generation_model = "offline-fixture"
            embedder = FixtureEmbedder()
            retriever = FixtureRetriever()
            generator = FixtureGenerator()

            def build_context(self, retrieved_chunks):
                context = namespace["build_context"](self, retrieved_chunks)
                receipt = {
                    "candidate_id": "fixture-c109",
                    "selected_variant": "char_rrf",
                    "runtime_identity": {
                        "scope": "OFFLINE_SYNTHETIC_FIXTURE",
                        "pipeline_commit": commit,
                        "quota_source_receipt_sha256": pins["C108"][1],
                        "char_ranker_source_receipt_sha256": pins["C100"][1],
                    },
                    "question_sha256": sha(question.encode("utf-8")),
                    "ordered_chunk_ids": [item["chunk_id"] for item in retrieved_chunks],
                    "ordered_chunk_text_sha256": [
                        sha(item["text"].encode("utf-8")) for item in retrieved_chunks
                    ],
                    "rendered_context_sha256": sha(context.encode("utf-8")),
                }
                write_verified(receipt_path, encode(receipt))
                events.append("INPUT_RECEIPT_WRITTEN")
                return context

        output = FixturePipeline().ask(question, top_k=len(chunks))
        require(
            events == ["INPUT_RECEIPT_WRITTEN", "GENERATOR_VERIFIED_INPUT_RECEIPT"]
            and output["answer"] == abstention
            and output["from_cache"] is False
            and output["sources"] == output["retrieved_chunks"] == [],
            "C109_ABSTENTION_OR_RECEIPT_MISMATCH",
        )
        receipt_bytes = receipt_path.read_bytes()

    receipt_sha = sha(receipt_bytes)
    saved_receipt = folder / f"C109.R11_fixture_input_receipt.{receipt_sha}.json"
    write_verified(saved_receipt, receipt_bytes)

    result = {
        "cell": "C109",
        "roadmap_action": "R11",
        "roadmap_subtask": "R11.3",
        "status": "OFFLINE_QUOTA_CHAR_RRF_RECEIPT_CHAIN_PASSED",
        "scope": "PINNED_FUNCTIONS_AND_ASK_ON_SYNTHETIC_INPUTS",
        "input_file_hashes": {cell: pin[1] for cell, pin in pins.items()},
        "local_commit": commit,
        "upstream_source_leads": upstream_leads,
        "required_real_inputs": prior["required_runtime_inputs"],
        "quota_top10_weight_075": ranked[:10],
        "quota_top10_weight_025_control": reverse_weight[:10],
        "fused_top10_passed_to_generator": ordered_ids,
        "events": events,
        "input_receipt": {
            "path": str(saved_receipt.relative_to(root)),
            "file_bytes_sha256": receipt_sha,
        },
        "response_binding": {
            "input_receipt_file_bytes_sha256": receipt_sha,
            "answer_text_utf8_sha256": sha(output["answer"].encode("utf-8")),
        },
        "next_product_dependency": (
            "VERIFY_RUNTIME_PRODUCERS_OF_HYBRID100_LOCAL100_AND_INCUMBENT50"
        ),
        "product_runtime_binding_verified": False,
        "end_to_end_blocker": prior["end_to_end_blocker"],
        "rank_reconstruction_for_unverified_candidates": "DEFERRED",
        "target_case_rankings_reconstructed": 0,
        "new_api_calls": 0,
        "product_regression_tests_executed": 0,
        "historical_gate": prior["historical_gate"],
        "selected_version_gate": "BLOCKED",
    }

    data = encode(result)
    digest = sha(data)
    output_path = folder / f"C109.R11_offline_rank_and_receipt_chain.{digest}.json"
    write_verified(output_path, data)
    print(json.dumps({
        **{key: value for key, value in result.items() if key != "upstream_source_leads"},
        "upstream_source_lead_cell_indices": [
            item["cell_index_zero_based"] for item in upstream_leads
        ],
        "output": {
            "path": str(output_path.relative_to(root)),
            "file_bytes_sha256": digest,
            "reload_validation": "PASS",
        },
    }, ensure_ascii=False, indent=2))
    return result


V3_C109 = run_v3_c109()
````

````python
# C113 | R11.3 | Run after C001. No new imports.

def run_v3_c113():
    def sha(data):
        return hashlib.sha256(data).hexdigest()

    root = Path(os.environ.get(
        "RAG_FINANCE_ROOT",
        "/Users/macbookpro/AMIR_DATA/00_WORKS/01_PROJECTS/RAG-projects/RAG_finance",
    )).expanduser().resolve()
    folder = root / "reports/phase2/quality_improvement/NB22E.lab.v3"

    pins = {
        "C098": (
            "C098.R11_default_product_retriever_trace",
            "145964d199581d82acdbd2ccfd68610390838ed3722f96de367bb2b851546430",
        ),
        "C112": (
            "C112.R11_product_integration_acceptance_checks",
            "275e1e1968fa33beb35682c8795acc7638a27d0b1b476046320715f3c65e8a07",
        ),
    }
    documents = {}
    for cell, (stem, digest) in pins.items():
        raw = (folder / f"{stem}.{digest}.json").read_bytes()
        if sha(raw) != digest:
            raise ValueError(f"C113_INPUT_HASH_MISMATCH: {cell}")
        document = json.loads(raw)
        if document["cell"] != cell or document["roadmap_subtask"] != "R11.3":
            raise ValueError(f"C113_INPUT_IDENTITY_MISMATCH: {cell}")
        documents[cell] = document

    trace = documents["C098"]
    prior = documents["C112"]
    commit = trace["local_commit"]
    expected_files = {
        "src/pipeline/rag_pipeline.py",
        "src/retrieval/qdrant_retriever.py",
    }
    if (
        commit != "9bc25f986e0ccc0d83b2fde66ab583b8a64c4d18"
        or prior["local_commit"] != commit
        or prior["input_file_hashes"]["C098"] != pins["C098"][1]
        or prior["open_item_ids"] != [8, 9, 10, 11, 12]
        or prior["selected_version_gate"] != "BLOCKED"
        or set(trace["source_file_hashes"]) != expected_files
    ):
        raise ValueError("C113_PRIOR_BINDING_MISMATCH")

    comparisons = []
    for relative_path, expected_sha in sorted(trace["source_file_hashes"].items()):
        path = root / relative_path
        if not path.resolve().is_relative_to(root):
            raise ValueError(f"C113_SOURCE_PATH_OUTSIDE_PROJECT: {relative_path}")

        row = {"path": relative_path, "reference_sha256": expected_sha}
        if not path.is_file():
            row["status"] = "WORKTREE_FILE_UNAVAILABLE"
        else:
            actual_sha = sha(path.read_bytes())
            row["working_tree_sha256"] = actual_sha
            if actual_sha == expected_sha:
                row["status"] = "MATCHES_TRACED_DEFAULT_ROUTE"
            else:
                row["status"] = "CHANGED_REQUIRES_SOURCE_REVIEW"
                row["diff"] = subprocess.check_output(
                    [
                        "git", "-C", str(root), "diff", "--no-ext-diff",
                        "--no-textconv", commit, "--", relative_path,
                    ],
                    timeout=30,
                ).decode("utf-8", errors="replace")
        comparisons.append(row)

    unchanged = all(
        row["status"] == "MATCHES_TRACED_DEFAULT_ROUTE" for row in comparisons
    )
    result = {
        "cell": "C113",
        "roadmap_action": "R11",
        "roadmap_subtask": "R11.3",
        "status": (
            "TRACED_DEFAULT_ROUTE_STILL_MATCHES_WORKTREE"
            if unchanged else "WORKTREE_ROUTE_REQUIRES_TARGETED_REVIEW"
        ),
        "reference_commit": commit,
        "input_file_hashes": {cell: pin[1] for cell, pin in pins.items()},
        "source_comparisons": comparisons,
        "next_action": (
            "IMPLEMENT_OPT_IN_CHAR_RRF_WIRING_FOR_TRACED_DEFAULT_ROUTE"
            if unchanged else "REVIEW_CHANGED_OR_UNAVAILABLE_PRODUCT_FILES"
        ),
        "newly_completed_items_from_previous_open_set": [],
        "done_item_count": prior["done_item_count"],
        "open_item_count": prior["open_item_count"],
        "open_item_ids": prior["open_item_ids"],
        "product_runtime_binding_verified": False,
        "rank_reconstruction_for_unverified_candidates": "DEFERRED",
        "new_api_calls": 0,
        "product_regression_tests_executed": 0,
        "historical_gate": prior["historical_gate"],
        "selected_version_gate": "BLOCKED",
    }

    data = json.dumps(
        result, ensure_ascii=False, sort_keys=True, separators=(",", ":"),
        allow_nan=False,
    ).encode("utf-8")
    digest = sha(data)
    output = folder / f"C113.R11_product_route_worktree_check.{digest}.json"
    if not output.exists():
        with tempfile.NamedTemporaryFile(dir=folder, prefix=".C113-", delete=False) as handle:
            temporary = Path(handle.name)
            try:
                handle.write(data)
                handle.flush()
                os.fsync(handle.fileno())
                try:
                    os.link(temporary, output)
                except FileExistsError:
                    pass
            finally:
                temporary.unlink(missing_ok=True)

    if output.read_bytes() != data:
        raise ValueError("C113_OUTPUT_RELOAD_MISMATCH")

    print(json.dumps({
        **result,
        "output": {
            "path": str(output.relative_to(root)),
            "file_bytes_sha256": digest,
            "reload_validation": "PASS",
        },
    }, ensure_ascii=False, indent=2))
    return result


V3_C113 = run_v3_c113()
````

````python
# C114 | R11.3 | Run after C001. No new notebook imports.

def run_v3_c114():
    def sha(data):
        return hashlib.sha256(data).hexdigest()

    def require(condition, code):
        if not condition:
            raise ValueError(code)

    def encode(value):
        return json.dumps(
            value, ensure_ascii=False, sort_keys=True,
            separators=(",", ":"), allow_nan=False,
        ).encode("utf-8")

    def atomic_write(path, data):
        temporary = None
        try:
            with tempfile.NamedTemporaryFile(
                dir=path.parent, prefix=".C114-", delete=False,
            ) as handle:
                temporary = Path(handle.name)
                handle.write(data)
                handle.flush()
                os.fsync(handle.fileno())
            if path.exists():
                os.chmod(temporary, path.stat().st_mode & 0o777)
            os.replace(temporary, path)
        finally:
            if temporary is not None:
                temporary.unlink(missing_ok=True)
        require(path.read_bytes() == data, f"C114_RELOAD_MISMATCH: {path.name}")

    root = Path(os.environ.get(
        "RAG_FINANCE_ROOT",
        "/Users/macbookpro/AMIR_DATA/00_WORKS/01_PROJECTS/RAG-projects/RAG_finance",
    )).expanduser().resolve()
    folder = root / "reports/phase2/quality_improvement/NB22E.lab.v3"
    prior_sha = "4734551f64681f13f8553276783f27bde95a0212dbf401c16cc2c2152e777fe9"
    prior_path = folder / f"C113.R11_product_route_worktree_check.{prior_sha}.json"
    prior_raw = prior_path.read_bytes()
    require(sha(prior_raw) == prior_sha, "C114_C113_HASH_MISMATCH")
    prior = json.loads(prior_raw)
    require(
        prior["cell"] == "C113"
        and prior["status"] == "TRACED_DEFAULT_ROUTE_STILL_MATCHES_WORKTREE"
        and prior["next_action"] == "IMPLEMENT_OPT_IN_CHAR_RRF_WIRING_FOR_TRACED_DEFAULT_ROUTE"
        and prior["selected_version_gate"] == "BLOCKED",
        "C114_C113_STATE_MISMATCH",
    )

    relative_path = "src/pipeline/rag_pipeline.py"
    pipeline_path = root / relative_path
    expected = {
        row["path"]: row["reference_sha256"]
        for row in prior["source_comparisons"]
    }
    reference = subprocess.check_output(
        ["git", "-C", str(root), "show", f"{prior['reference_commit']}:{relative_path}"],
        timeout=30,
    )
    require(sha(reference) == expected[relative_path], "C114_REFERENCE_HASH_MISMATCH")

    retriever_path = "src/retrieval/qdrant_retriever.py"
    require(
        sha((root / retriever_path).read_bytes()) == expected[retriever_path],
        "C114_DEFAULT_RETRIEVER_CHANGED",
    )

    method_source = '''
    def ask_with_char_rrf(
        self, question, *, candidate_id, rank_sources, write_receipt, top_k=10
    ):
        """Run an explicitly supplied char_rrf route and persist its input receipt."""
        import hashlib

        def text_sha(text):
            return hashlib.sha256(text.encode("utf-8")).hexdigest()

        if not isinstance(question, str) or not question.strip():
            raise ValueError("CHAR_RRF_QUESTION_REQUIRED")
        if not isinstance(candidate_id, str) or not candidate_id.strip():
            raise ValueError("CHAR_RRF_CANDIDATE_ID_REQUIRED")
        if type(top_k) is not int or top_k < 1:
            raise ValueError("CHAR_RRF_INVALID_TOP_K")

        ranked = rank_sources(question=question, candidate_id=candidate_id, top_k=top_k)
        if (
            ranked["selected_variant"] != "char_rrf"
            or ranked["candidate_id"] != candidate_id
            or ranked["question_sha256"] != text_sha(question)
            or not isinstance(ranked["runtime_identity"], dict)
            or not ranked["runtime_identity"]
        ):
            raise ValueError("CHAR_RRF_RANK_RECEIPT_IDENTITY_MISMATCH")

        chunks = ranked["ordered_chunks"]
        if not isinstance(chunks, list) or not 0 < len(chunks) <= top_k:
            raise ValueError("CHAR_RRF_INVALID_CHUNK_COUNT")
        ids = [chunk["chunk_id"] for chunk in chunks]
        if any(not isinstance(value, str) or not value for value in ids):
            raise ValueError("CHAR_RRF_INVALID_CHUNK_ID")
        if len(set(ids)) != len(ids):
            raise ValueError("CHAR_RRF_DUPLICATE_CHUNK_ID")

        context = self.build_context(chunks)
        input_receipt = {
            "candidate_id": candidate_id,
            "question_sha256": text_sha(question),
            "selected_variant": "char_rrf",
            "runtime_identity": ranked["runtime_identity"],
            "ordered_chunk_ids": ids,
            "ordered_chunk_text_sha256": [text_sha(chunk["text"]) for chunk in chunks],
            "rendered_context_sha256": text_sha(context),
        }
        input_sha = write_receipt("input", input_receipt)
        if (
            not isinstance(input_sha, str)
            or len(input_sha) != 64
            or any(char not in "0123456789abcdef" for char in input_sha)
        ):
            raise ValueError("CHAR_RRF_INPUT_RECEIPT_HASH_REQUIRED")

        answer = self.generator.generate_answer(question=question, context=context)
        output_sha = write_receipt("output", {
            "input_receipt_file_bytes_sha256": input_sha,
            "answer_text_utf8_sha256": text_sha(answer),
        })
        return {
            "question": question,
            "answer": answer,
            "sources": [] if answer.strip() == "در منابع موجود پاسخ کافی پیدا نشد."
                       else self.format_sources(chunks),
            "selected_variant": "char_rrf",
            "input_receipt_file_bytes_sha256": input_sha,
            "output_receipt_file_bytes_sha256": output_sha,
            "from_cache": False,
        }

'''
    original = reference.decode("utf-8")
    anchor = "    def close(self):"
    require(original.count(anchor) == 1, "C114_INSERTION_POINT_NOT_UNIQUE")
    patched = original.replace(anchor, method_source + anchor, 1).encode("utf-8")
    working = pipeline_path.read_bytes()
    require(working in (reference, patched), "C114_PIPELINE_HAS_UNEXPECTED_CHANGES")

    # Compile the full proposed file. Execute only the fixture methods.
    patched_source = patched.decode("utf-8")
    compile(patched_source, str(pipeline_path), "exec")
    tree = ast.parse(patched_source)
    pipeline_class = next(
        node for node in tree.body
        if isinstance(node, ast.ClassDef) and node.name == "RAGPipeline"
    )
    method_names = {"ask_with_char_rrf", "build_context", "format_sources"}
    fixture_methods = [
        node for node in pipeline_class.body
        if isinstance(node, ast.FunctionDef) and node.name in method_names
    ]
    require(
        len(fixture_methods) == 3
        and {node.name for node in fixture_methods} == method_names,
        "C114_FIXTURE_METHOD_SET_MISMATCH",
    )
    namespace = {}
    exec(compile(ast.Module(body=fixture_methods, type_ignores=[]),
                 "<C114:fixture>", "exec"), namespace)

    question = "سؤال آزمایشی C114"
    candidate_id = "fixture-c114"
    chunks = [
        {
            "chunk_id": chunk_id, "title": chunk_id,
            "url": f"fixture://{chunk_id}", "text": f"متن {chunk_id}",
        }
        for chunk_id in ("fixture-b", "fixture-a")
    ]
    events = []
    with tempfile.TemporaryDirectory(dir=folder, prefix=".C114-fixture-") as work:
        work_path = Path(work)

        def write_receipt(kind, record):
            path = work_path / f"{kind}.json"
            data = encode(record)
            atomic_write(path, data)
            events.append(f"{kind}_receipt_written")
            return sha(data)

        def rank_sources(*, question, candidate_id, top_k):
            require(top_k == 2, "C114_FIXTURE_TOP_K_MISMATCH")
            return {
                "selected_variant": "char_rrf",
                "candidate_id": candidate_id,
                "question_sha256": sha(question.encode("utf-8")),
                "runtime_identity": {"scope": "SYNTHETIC_FIXTURE"},
                "ordered_chunks": chunks,
            }

        class FixtureGenerator:
            def generate_answer(self, *, question, context):
                require(events == ["input_receipt_written"], "C114_RECEIPT_TIMING_MISMATCH")
                receipt = json.loads((work_path / "input.json").read_bytes())
                require(
                    receipt["ordered_chunk_ids"] == ["fixture-b", "fixture-a"]
                    and context.index("شناسه چانک: fixture-b")
                    < context.index("شناسه چانک: fixture-a")
                    and receipt["rendered_context_sha256"] == sha(context.encode("utf-8")),
                    "C114_FIXTURE_CONTEXT_MISMATCH",
                )
                events.append("generator_called")
                return "در منابع موجود پاسخ کافی پیدا نشد."

        fixture = type("FixturePipeline", (), namespace)()
        fixture.generator = FixtureGenerator()
        answer = fixture.ask_with_char_rrf(
            question, candidate_id=candidate_id, rank_sources=rank_sources,
            write_receipt=write_receipt, top_k=2,
        )
        output_receipt = json.loads((work_path / "output.json").read_bytes())
        require(
            events == ["input_receipt_written", "generator_called", "output_receipt_written"]
            and answer["sources"] == []
            and output_receipt["input_receipt_file_bytes_sha256"]
            == sha((work_path / "input.json").read_bytes())
            == answer["input_receipt_file_bytes_sha256"]
            and output_receipt["answer_text_utf8_sha256"]
            == sha(answer["answer"].encode("utf-8")),
            "C114_FIXTURE_RESPONSE_BINDING_MISMATCH",
        )

    require(pipeline_path.read_bytes() == working, "C114_PIPELINE_CHANGED_DURING_CHECK")
    if working == reference:
        atomic_write(pipeline_path, patched)

    result = {
        "cell": "C114",
        "roadmap_subtask": "R11.3",
        "status": "OPT_IN_CHAR_RRF_ENTRYPOINT_WRITTEN_AND_OFFLINE_FIXTURE_PASSED",
        "pipeline_path": relative_path,
        "before_sha256": sha(reference),
        "after_sha256": sha(patched),
        "fixture_events": events,
        "remaining_item_11_dependency": "BIND_REAL_VERSIONED_QUOTA100_AND_CHAR_RRF_PROVIDER",
        "done_item_count": 7,
        "open_item_count": 5,
        "newly_completed_items_from_previous_open_set": [],
        "product_runtime_binding_verified": False,
        "new_api_calls": 0,
        "product_regression_tests_executed": 0,
        "selected_version_gate": "BLOCKED",
    }
    data = encode(result)
    output = folder / f"C114.R11_opt_in_char_rrf_entrypoint.{sha(data)}.json"
    atomic_write(output, data)
    print(json.dumps({
        **result,
        "output": str(output.relative_to(root)),
    }, ensure_ascii=False, indent=2))
    return result


V3_C114 = run_v3_c114()
````

````python
# C118 | R11.3 | Run after C001. No new imports.

def run_v3_c118():
    def sha(data):
        return hashlib.sha256(data).hexdigest()

    def require(condition, code):
        if not condition:
            raise ValueError(code)

    root = Path(os.environ.get(
        "RAG_FINANCE_ROOT",
        "/Users/macbookpro/AMIR_DATA/00_WORKS/01_PROJECTS/RAG-projects/RAG_finance",
    )).expanduser().resolve()
    folder = root / "reports/phase2/quality_improvement/NB22E.lab.v3"
    c117_sha = "e326fb28516942a5b9d35971ee6158a450d9bf13465bf7faef336d511c56270f"
    c117_path = folder / f"C117.R11_nb22b_runtime_input_implementation.{c117_sha}.json"
    raw = c117_path.read_bytes()
    require(sha(raw) == c117_sha, "C118_C117_HASH_MISMATCH")
    prior = json.loads(raw)
    require(
        prior["cell"] == "C117"
        and prior["roadmap_subtask"] == "R11.3"
        and prior["selected_version_gate"] == "BLOCKED"
        and prior["product_runtime_binding_verified"] is False,
        "C118_C117_STATE_MISMATCH",
    )

    notebook_path = "notebooks/phase2/22b_quality_improvement_and_acceptance.ipynb"
    commit = prior["local_commit"]
    notebook_bytes = subprocess.check_output(
        ["git", "-C", str(root), "show", f"{commit}:{notebook_path}"],
        timeout=60,
    )
    notebook_sha = sha(notebook_bytes)
    require(
        notebook_sha == prior["input_file_hashes"]["NB22B"],
        "C118_NB22B_HASH_MISMATCH",
    )

    required_keys = {
        "model_id", "model_revision", "onnx_sha256", "onnxruntime_version",
        "batch_size", "maximum_sequence_length",
    }
    config_matches = []
    for index, cell in enumerate(json.loads(notebook_bytes)["cells"]):
        if cell["cell_type"] != "code":
            continue
        source = cell.get("source", "")
        source = source if isinstance(source, str) else "".join(source)
        if "CONFIG" not in source:
            continue
        try:
            tree = ast.parse(source)
        except SyntaxError:
            continue
        for node in tree.body:
            if not isinstance(node, ast.Assign) or not any(
                isinstance(target, ast.Name) and target.id == "CONFIG"
                for target in node.targets
            ):
                continue
            try:
                value = ast.literal_eval(node.value)
            except (ValueError, TypeError):
                continue
            if isinstance(value, dict) and required_keys <= value.keys():
                config_matches.append((index, value))

    require(len(config_matches) == 1, "C118_RUNTIME_CONFIG_NOT_UNIQUE")
    config_index, config = config_matches[0]
    require(
        config["batch_size"] == 4 and config["maximum_sequence_length"] == 512,
        "C118_PREPROCESSING_CONTRACT_MISMATCH",
    )
    model_sha = config["onnx_sha256"]
    require(
        isinstance(model_sha, str) and len(model_sha) == 64
        and all(char in "0123456789abcdef" for char in model_sha),
        "C118_INVALID_PINNED_MODEL_HASH",
    )

    # Same local cache resolution as the recorded local_depth100.
    cache_root = Path(os.environ.get("HF_HUB_CACHE") or str(
        Path(os.environ.get("HF_HOME", str(Path.home() / ".cache/huggingface"))) / "hub"
    ))
    snapshot = (
        cache_root / ("models--" + config["model_id"].replace("/", "--"))
        / "snapshots" / config["model_revision"]
    )
    model = snapshot / "onnx/model_quint8_avx2.onnx"

    actual_model_sha = None
    if model.is_file():
        digest = hashlib.sha256()
        with model.open("rb") as handle:
            while block := handle.read(4 * 1024 * 1024):
                digest.update(block)
        actual_model_sha = digest.hexdigest()

    model_status = (
        "FILE_UNAVAILABLE" if actual_model_sha is None
        else "FILE_HASH_MATCHED" if actual_model_sha == model_sha
        else "FILE_HASH_MISMATCH"
    )

    tokenizer_files = [
        name for name in (
            "config.json", "tokenizer.json", "tokenizer_config.json",
            "special_tokens_map.json", "sentencepiece.bpe.model",
            "tokenizer.model", "vocab.txt", "vocab.json", "merges.txt",
        )
        if (snapshot / name).is_file()
    ]

    result = {
        "cell": "C118",
        "roadmap_action": "R11",
        "roadmap_subtask": "R11.3",
        "status": "LOCAL100_PINNED_MODEL_PREFLIGHT_RECORDED",
        "input_file_hashes": {"C117": c117_sha, "NB22B": notebook_sha},
        "local_commit": commit,
        "configuration_cell_index_zero_based": config_index,
        "expected_runtime": {key: config[key] for key in sorted(required_keys)},
        "snapshot_exists": snapshot.is_dir(),
        "onnx_model_status": model_status,
        "actual_onnx_sha256": actual_model_sha,
        "tokenizer_files_present": tokenizer_files,
        "tokenizer_loadability_verified": False,
        "installed_onnxruntime_version_verified": False,
        "score_parity_verified": False,
        "next_local100_dependency": (
            "VERIFY_LOCAL_RUNTIME_TOKENIZER_AND_FOUR_PAIR_PARITY"
            if model_status == "FILE_HASH_MATCHED"
            else "RESOLVE_PINNED_ONNX_FILE_OR_USE_VERIFIED_COMPLETE_CACHE"
        ),
        "other_product_input_dependency": "BIND_SAME_QUERY_HYBRID100_AND_SCORED_POOL50",
        "done_item_count": prior["done_item_count"],
        "open_item_count": prior["open_item_count"],
        "newly_completed_items_from_previous_open_set": [],
        "product_runtime_binding_verified": False,
        "rank_reconstruction_for_unverified_candidates": "DEFERRED",
        "new_api_calls": 0,
        "model_inference_calls": 0,
        "product_regression_tests_executed": 0,
        "historical_gate": prior["historical_gate"],
        "selected_version_gate": "BLOCKED",
    }

    data = json.dumps(
        result, ensure_ascii=False, sort_keys=True, separators=(",", ":"),
        allow_nan=False,
    ).encode("utf-8")
    digest = sha(data)
    output = folder / f"C118.R11_local100_model_preflight.{digest}.json"
    if not output.exists():
        temporary = None
        try:
            with tempfile.NamedTemporaryFile(
                dir=folder, prefix=".C118-", delete=False,
            ) as handle:
                temporary = Path(handle.name)
                handle.write(data)
                handle.flush()
                os.fsync(handle.fileno())
            try:
                os.link(temporary, output)
            except FileExistsError:
                pass
        finally:
            if temporary is not None:
                temporary.unlink(missing_ok=True)

    require(output.read_bytes() == data, "C118_OUTPUT_RELOAD_MISMATCH")
    print(json.dumps({
        **result,
        "output": {
            "path": str(output.relative_to(root)),
            "file_bytes_sha256": digest,
            "reload_validation": "PASS",
        },
    }, ensure_ascii=False, indent=2))
    return result


V3_C118 = run_v3_c118()
````

### سل ۱۱۹ — اتصال ورودی‌های محلی و کنترل برابری چهار امتیاز

رسید سل ۱۱۸ و چهار فایل ورودی گزارش ۰۵۶ با هش کنترل می‌شوند. اولین سؤال Development به ترتیب شناسه، مطابق تهیه‌کنندهٔ اصلی انتخاب می‌شود. عمق و ترتیب Hybrid100 و Pool50 همان سؤال کنترل می‌شود. سپس مدل و توکنایزر پین‌شده از کش محلی بار می‌شوند و یک دستهٔ چهارتایی روی CPU امتیاز می‌گیرد.

حد اختلاف مجاز با امتیازهای ذخیره‌شده `0.0001` است. وضعیت ناموفق همراه مرحلهٔ دقیق ثبت می‌شود. وضعیت چک‌لیست اجرایی ۷ انجام‌شده و ۵ باز است. اجرای این سل آفلاین است و بازسازی ۱۶ رتبه همچنان معوق است.

</div>

````python
# C119 — R11.3 — Offline four-pair score parity
def run_v3_c119():
    def require(condition, code):
        if not condition:
            raise ValueError(code)

    def sha(raw):
        return hashlib.sha256(raw).hexdigest()

    reports = Path("reports/phase2/quality_improvement/NB22E.lab.v3")
    root = next(
        (p for p in (Path.cwd(), *Path.cwd().parents)
         if (p / reports).is_dir()),
        None,
    )
    require(root is not None, "C119_PROJECT_ROOT_NOT_FOUND")
    root = root.resolve()

    def read_bound(relative, expected, *, rows=False):
        path = (root / relative).resolve()
        require(path.is_relative_to(root), "C119_PATH_OUTSIDE_PROJECT")
        raw = path.read_bytes()
        require(sha(raw) == expected, f"C119_HASH_CONFLICT:{relative}")
        if rows:
            return [json.loads(line) for line in raw.splitlines() if line.strip()]
        return json.loads(raw)

    def index_rows(rows, field):
        indexed = {}
        for row in rows:
            key = row[field]
            require(isinstance(key, str) and bool(key), f"C119_INVALID_ID:{field}")
            require(key not in indexed, f"C119_DUPLICATE_ID:{field}:{key}")
            indexed[key] = row
        return indexed

    def rank_rows(row, depth):
        records = row["candidates"]
        require(len(records) == depth, f"C119_DEPTH_CONFLICT:{depth}")
        require(
            [r["rank"] for r in records] == list(range(1, depth + 1)),
            f"C119_RANK_ORDER_CONFLICT:{depth}",
        )
        return index_rows(records, "chunk_id")

    preflight_sha = "21aee5d3199de8d81afc83b58f96ff67df21fc47b019132c52b8d67235524c4d"
    source_sha = "ff6d6e136043929e0156e43f6ea75e65bd64f40ab7bcdafef90a4ba40901de22"
    c118 = read_bound(
        reports / f"C118.R11_local100_model_preflight.{preflight_sha}.json",
        preflight_sha,
    )
    require(c118["onnx_model_status"] == "FILE_HASH_MATCHED", "C119_PREFLIGHT_REQUIRED")
    expected = c118["expected_runtime"]

    result = {
        "cell": "C119",
        "roadmap_action": "R11",
        "roadmap_subtask": "R11.3",
        "input_file_hashes": {"C118": preflight_sha, "NB22_C056": source_sha},
        "scope": "FIRST_FOUR_PINNED_DEVELOPMENT_HYBRID_PAIRS_CPU_ONLY",
        "score_parity_verified": False,
        "tokenizer_loadability_verified": False,
        "installed_onnxruntime_version_verified": False,
        "same_query_hybrid100_and_scored_pool50_bound": False,
        "model_inference_calls": 0,
        "new_api_calls": 0,
        "target_case_rankings_reconstructed": 0,
        "rank_reconstruction_for_unverified_candidates": "DEFERRED",
        "product_regression_tests_executed": 0,
        "product_runtime_binding_verified": False,
        "done_item_count": 7,
        "open_item_count": 5,
        "newly_completed_items_from_previous_open_set": [],
        "historical_gate": "FAIL",
        "selected_version_gate": "BLOCKED",
    }

    stage = "PINNED_INPUT_BINDING"
    try:
        c056 = read_bound(
            Path("reports/phase2/release/NB22.C056.retrieval_verification")
            / f"C056.independent_retrieval_metric_verification.{source_sha}.json",
            source_sha,
        )
        require(
            c056["independent_recomputation_complete"] is True
            and c056["all_recorded_metric_comparisons_match"] is True,
            "C119_C056_VERIFICATION_REQUIRED",
        )
        inputs = {}
        for role, key in (
            ("golden", "candidate_id"),
            ("hybrid", "candidate_id"),
            ("reranker", "candidate_id"),
            ("chunks", "chunk_id"),
        ):
            receipt = c056["input_file_receipts"][role]
            relative = Path(receipt["path"])
            require(not relative.is_absolute(), f"C119_ABSOLUTE_INPUT_PATH:{role}")
            digest = receipt["file_bytes_sha256"]
            inputs[role] = index_rows(read_bound(relative, digest, rows=True), key)
            result["input_file_hashes"][role] = digest

        require(
            set(inputs["golden"]) == set(inputs["hybrid"]) == set(inputs["reranker"]),
            "C119_QUERY_POPULATION_CONFLICT",
        )
        development_ids = sorted(
            qid for qid, row in inputs["hybrid"].items()
            if row["split"] == "development"
        )
        require(len(development_ids) == 2664, "C119_DEVELOPMENT_POPULATION_CONFLICT")
        candidate_id = development_ids[0]
        hybrid = rank_rows(inputs["hybrid"][candidate_id], 100)
        reranker = rank_rows(inputs["reranker"][candidate_id], 50)
        hybrid_ids = list(hybrid)
        require(set(hybrid_ids[:50]) == set(reranker), "C119_POOL50_MEMBERSHIP_CONFLICT")
        require(set(hybrid_ids) <= set(inputs["chunks"]), "C119_UNKNOWN_HYBRID_CHUNK")

        question = inputs["golden"][candidate_id]["evaluation_content"]["question"].strip()
        require(bool(question), "C119_EMPTY_QUESTION")
        pair_ids = hybrid_ids[:4]
        contents = [inputs["chunks"][cid]["content"] for cid in pair_ids]
        reference_scores = [float(reranker[cid]["reranker_score"]) for cid in pair_ids]
        result.update(
            sample_candidate_id=candidate_id,
            question_sha256=sha(question.encode("utf-8")),
            pair_chunk_ids=pair_ids,
            pair_content_sha256=[sha(value.encode("utf-8")) for value in contents],
            reference_scores=reference_scores,
            same_query_hybrid100_and_scored_pool50_bound=True,
        )

        stage = "C001_RUNTIME_BINDINGS"
        namespace = globals()
        runtime = {
            "numpy": namespace.get("np"),
            "onnxruntime": namespace.get("ort") or namespace.get("onnxruntime"),
            "AutoTokenizer": namespace.get("AutoTokenizer"),
        }
        missing = [name for name, value in runtime.items() if value is None]
        result["missing_c001_runtime_bindings"] = missing
        require(not missing, "C119_C001_RUNTIME_BINDINGS_MISSING")
        numpy = runtime["numpy"]
        ort_local = runtime["onnxruntime"]

        stage = "PINNED_RUNTIME_VERSION"
        result["installed_onnxruntime_version"] = ort_local.__version__
        require(
            ort_local.__version__ == expected["onnxruntime_version"],
            "C119_ONNXRUNTIME_VERSION_CONFLICT",
        )
        result["installed_onnxruntime_version_verified"] = True

        stage = "LOCAL_SNAPSHOT_AND_TOKENIZER"
        cache_root = Path(
            os.environ.get("HF_HUB_CACHE")
            or str(Path(os.environ.get("HF_HOME", str(Path.home() / ".cache/huggingface"))) / "hub")
        )
        snapshot = (
            cache_root / ("models--" + expected["model_id"].replace("/", "--"))
            / "snapshots" / expected["model_revision"]
        )
        model = snapshot / "onnx/model_quint8_avx2.onnx"
        require(sha(model.read_bytes()) == expected["onnx_sha256"], "C119_MODEL_HASH_CONFLICT")
        tokenizer = runtime["AutoTokenizer"].from_pretrained(
            str(snapshot), local_files_only=True, trust_remote_code=False
        )
        result["tokenizer_loadability_verified"] = True

        stage = "CPU_SESSION_AND_INPUT_CONTRACT"
        require(expected["batch_size"] == 4, "C119_BATCH_SIZE_CONFLICT")
        require(expected["maximum_sequence_length"] == 512, "C119_SEQUENCE_LENGTH_CONFLICT")
        options = ort_local.SessionOptions()
        options.intra_op_num_threads = 4
        options.inter_op_num_threads = 1
        options.execution_mode = ort_local.ExecutionMode.ORT_SEQUENTIAL
        options.graph_optimization_level = ort_local.GraphOptimizationLevel.ORT_ENABLE_ALL
        session = ort_local.InferenceSession(
            str(model), sess_options=options, providers=["CPUExecutionProvider"]
        )
        names = {item.name for item in session.get_inputs()}
        require({"input_ids", "attention_mask"} <= names, "C119_ONNX_INPUT_CONTRACT")
        encoded = tokenizer(
            [question] * 4, contents, padding=True, truncation=True,
            max_length=512, return_tensors="np",
        )
        require(names <= set(encoded), "C119_TOKENIZER_ONNX_INPUT_MISMATCH")
        feed = {name: numpy.asarray(encoded[name], dtype=numpy.int64) for name in names}

        stage = "FOUR_PAIR_SCORE_PARITY"
        require(bool(numpy.isfinite(reference_scores).all()), "C119_INVALID_REFERENCE_SCORES")
        result["model_inference_calls"] += 1
        scores = numpy.asarray(session.run(None, feed)[0], dtype=numpy.float64).reshape(-1)
        require(scores.shape == (4,) and bool(numpy.isfinite(scores).all()), "C119_INVALID_ONNX_SCORES")
        deltas = [abs(float(score) - ref) for score, ref in zip(scores, reference_scores)]
        result.update(
            observed_scores=scores.tolist(),
            absolute_deltas=deltas,
            maximum_absolute_delta=max(deltas),
            maximum_allowed_delta=1e-4,
        )
        require(max(deltas) <= 1e-4, "C119_SCORE_PARITY_NOT_ESTABLISHED")
        result.update(
            status="PINNED_LOCAL_RUNTIME_FOUR_PAIR_PARITY_PASSED",
            score_parity_verified=True,
            blocker=None,
            next_product_dependency="BIND_VERSIONED_PROVIDERS_AT_OPT_IN_PRODUCT_ENTRYPOINT",
        )
    except Exception as error:
        result.update(
            status="LOCAL_RUNTIME_PARITY_BLOCKED",
            blocker={"stage": stage, "type": type(error).__name__, "detail": str(error)},
        )

    raw = json.dumps(
        result, ensure_ascii=False, sort_keys=True, indent=2, allow_nan=False
    ).encode("utf-8")
    digest = sha(raw)
    relative = reports / f"C119.R11_local100_four_pair_parity.{digest}.json"
    path = root / relative
    try:
        with path.open("xb") as handle:
            handle.write(raw)
            handle.flush()
            os.fsync(handle.fileno())
    except FileExistsError:
        pass
    require(path.read_bytes() == raw, "C119_REPORT_RELOAD_CONFLICT")
    result["output"] = {
        "path": relative.as_posix(),
        "file_bytes_sha256": digest,
        "reload_validation": "PASS",
    }
    print(json.dumps(result, ensure_ascii=False, indent=2))
    return result


V3_C119 = run_v3_c119()
````

### سل ۱۲۰ — ساخت ترتیب Local100 برای یک سؤال Development

رسید قبولی سل ۱۱۹ مصرف می‌شود. ترتیب Hybrid100 و امتیازهای Pool50 از فایل‌های پین‌شدهٔ گزارش ۰۵۶ خوانده می‌شوند. برای همان سؤال، مدل محلی فقط ۵۰ جفت فاقد امتیاز را پردازش می‌کند. ترتیب Local100 بر اساس امتیاز نزولی و شناسهٔ قطعه در تساوی ساخته و با هش ورودی‌ها ثبت می‌شود.

دامنه یک سؤال Development است. اجرای آفلاین، بدون فراخوانی API و بدون بازسازی رتبه‌های ۱۶ پروندهٔ معوق انجام می‌شود.

</div>

````python
# C120 — R11.3 — Build one version-bound Development Local100 input
def run_v3_c120():
    def require(ok, code):
        if not ok:
            raise ValueError(code)

    def sha(raw):
        return hashlib.sha256(raw).hexdigest()

    reports = Path("reports/phase2/quality_improvement/NB22E.lab.v3")
    root = next(
        (p for p in (Path.cwd(), *Path.cwd().parents) if (p / reports).is_dir()),
        None,
    )
    require(root is not None, "C120_PROJECT_ROOT_NOT_FOUND")
    root = root.resolve()

    def read_bound(relative, digest, *, key=None):
        relative = Path(relative)
        path = (root / relative).resolve()
        require(
            not relative.is_absolute() and path.is_relative_to(root),
            "C120_PATH_OUTSIDE_PROJECT",
        )
        raw = path.read_bytes()
        require(sha(raw) == digest, f"C120_HASH_CONFLICT:{relative}")
        if key is None:
            return json.loads(raw)
        rows = {}
        for line in raw.splitlines():
            if line.strip():
                row = json.loads(line)
                identity = row[key]
                require(identity not in rows, f"C120_DUPLICATE_ID:{relative}:{identity}")
                rows[identity] = row
        return rows

    def ordered_ids(row, depth):
        candidates = row["candidates"]
        require(len(candidates) == depth, f"C120_DEPTH_CONFLICT:{depth}")
        require(
            [r["rank"] for r in candidates] == list(range(1, depth + 1)),
            f"C120_RANK_ORDER_CONFLICT:{depth}",
        )
        ids = [r["chunk_id"] for r in candidates]
        require(len(set(ids)) == depth, f"C120_DUPLICATE_CHUNK:{depth}")
        return ids

    c119_sha = "967178107d6a4fab03ecb5ffd74f79ceebfd716dc265d9726339164d06a3647b"
    prior = read_bound(
        reports / f"C119.R11_local100_four_pair_parity.{c119_sha}.json",
        c119_sha,
    )
    require(
        prior["score_parity_verified"] is True
        and prior["same_query_hybrid100_and_scored_pool50_bound"] is True,
        "C120_C119_PARITY_REQUIRED",
    )

    pins = prior["input_file_hashes"]
    candidate = prior["sample_candidate_id"]
    result = {
        "cell": "C120",
        "roadmap_action": "R11",
        "roadmap_subtask": "R11.3",
        "scope": "ONE_DEVELOPMENT_QUERY_PINNED_STORED_INPUTS",
        "candidate_id": candidate,
        "input_file_hashes": {"C119": c119_sha, **pins},
        "new_api_calls": 0,
        "model_inference_calls_attempted": 0,
        "new_local_pairs_scored": 0,
        "target_case_rankings_reconstructed": 0,
        "product_runtime_binding_verified": False,
        "product_regression_tests_executed": 0,
        "done_item_count": 7,
        "open_item_count": 5,
        "newly_completed_items_from_previous_open_set": [],
        "historical_gate": "FAIL",
        "selected_version_gate": "BLOCKED",
    }

    stage = "PINNED_INPUTS"
    try:
        c056_sha = pins["NB22_C056"]
        c056 = read_bound(
            Path("reports/phase2/release/NB22.C056.retrieval_verification")
            / f"C056.independent_retrieval_metric_verification.{c056_sha}.json",
            c056_sha,
        )
        inputs = {}
        for role, key in (
            ("golden", "candidate_id"),
            ("hybrid", "candidate_id"),
            ("reranker", "candidate_id"),
            ("chunks", "chunk_id"),
        ):
            receipt = c056["input_file_receipts"][role]
            require(receipt["file_bytes_sha256"] == pins[role], f"C120_PIN_CONFLICT:{role}")
            inputs[role] = read_bound(receipt["path"], pins[role], key=key)

        hrow = inputs["hybrid"][candidate]
        rrow = inputs["reranker"][candidate]
        require(hrow["split"] == "development", "C120_DEVELOPMENT_REQUIRED")
        hybrid_ids = ordered_ids(hrow, 100)
        pool_ids = ordered_ids(rrow, 50)
        require(set(hybrid_ids[:50]) == set(pool_ids), "C120_POOL50_MEMBERSHIP_CONFLICT")

        question = inputs["golden"][candidate]["evaluation_content"]["question"].strip()
        question_sha = sha(question.encode("utf-8"))
        require(question_sha == prior["question_sha256"], "C120_QUESTION_HASH_CONFLICT")
        require(hybrid_ids[:4] == prior["pair_chunk_ids"], "C120_PARITY_PAIR_ID_CONFLICT")

        contents = {cid: inputs["chunks"][cid]["content"] for cid in hybrid_ids}
        content_hashes = {cid: sha(value.encode("utf-8")) for cid, value in contents.items()}
        require(
            [content_hashes[cid] for cid in hybrid_ids[:4]] == prior["pair_content_sha256"],
            "C120_PARITY_PAIR_TEXT_CONFLICT",
        )
        scores = {
            row["chunk_id"]: float(row["reranker_score"]) for row in rrow["candidates"]
        }
        require(
            [scores[cid] for cid in hybrid_ids[:4]] == prior["reference_scores"],
            "C120_PARITY_REFERENCE_SCORE_CONFLICT",
        )
        require(bool(np.isfinite(list(scores.values())).all()), "C120_INVALID_POOL50_SCORE")

        stage = "PINNED_LOCAL_RUNTIME"
        c118_sha = pins["C118"]
        preflight = read_bound(
            reports / f"C118.R11_local100_model_preflight.{c118_sha}.json",
            c118_sha,
        )
        config = preflight["expected_runtime"]
        runtime = globals().get("ort") or globals().get("onnxruntime")
        require(runtime is not None, "C120_C001_ONNXRUNTIME_BINDING_MISSING")
        require(
            runtime.__version__ == config["onnxruntime_version"],
            "C120_ONNXRUNTIME_VERSION_CONFLICT",
        )
        require(
            (config["batch_size"], config["maximum_sequence_length"]) == (4, 512),
            "C120_INFERENCE_CONFIG_CONFLICT",
        )
        cache_root = Path(
            os.environ.get("HF_HUB_CACHE")
            or str(Path(os.environ.get("HF_HOME", str(Path.home() / ".cache/huggingface"))) / "hub")
        )
        snapshot = (
            cache_root / ("models--" + config["model_id"].replace("/", "--"))
            / "snapshots" / config["model_revision"]
        )
        model = snapshot / "onnx/model_quint8_avx2.onnx"
        require(sha(model.read_bytes()) == config["onnx_sha256"], "C120_MODEL_HASH_CONFLICT")
        tokenizer = AutoTokenizer.from_pretrained(
            str(snapshot), local_files_only=True, trust_remote_code=False
        )
        options = runtime.SessionOptions()
        options.intra_op_num_threads = 4
        options.inter_op_num_threads = 1
        options.execution_mode = runtime.ExecutionMode.ORT_SEQUENTIAL
        options.graph_optimization_level = runtime.GraphOptimizationLevel.ORT_ENABLE_ALL
        session = runtime.InferenceSession(
            str(model), sess_options=options, providers=["CPUExecutionProvider"]
        )
        names = {item.name for item in session.get_inputs()}
        require({"input_ids", "attention_mask"} <= names, "C120_ONNX_INPUT_CONTRACT")

        stage = "SCORE_MISSING_50"
        missing = [cid for cid in hybrid_ids if cid not in scores]
        require(len(missing) == 50, "C120_MISSING_PAIR_COUNT_CONFLICT")
        for start in range(0, len(missing), 4):
            batch = missing[start:start + 4]
            encoded = tokenizer(
                [question] * len(batch), [contents[cid] for cid in batch],
                padding=True, truncation=True, max_length=512, return_tensors="np",
            )
            require(names <= set(encoded), "C120_TOKENIZER_ONNX_INPUT_MISMATCH")
            feed = {name: np.asarray(encoded[name], dtype=np.int64) for name in names}
            result["model_inference_calls_attempted"] += 1
            values = np.asarray(session.run(None, feed)[0], dtype=np.float64).reshape(-1)
            require(
                values.shape == (len(batch),) and bool(np.isfinite(values).all()),
                "C120_INVALID_ONNX_SCORES",
            )
            scores.update({cid: float(value) for cid, value in zip(batch, values)})
            result["new_local_pairs_scored"] += len(batch)

        require(set(scores) == set(hybrid_ids), "C120_LOCAL100_CHUNK_SET_CONFLICT")
        local_ids = sorted(hybrid_ids, key=lambda cid: (-scores[cid], cid))
        result.update(
            status="ONE_DEVELOPMENT_LOCAL100_INPUT_BUILT",
            blocker=None,
            question_sha256=question_sha,
            hybrid100_order=hybrid_ids,
            scored_pool50_order=pool_ids,
            local100_order=local_ids,
            scores_by_chunk_id=scores,
            chunk_text_sha256_by_id=content_hashes,
            reused_pool50_score_count=50,
            local100_matches_saved_order_verified=False,
            next_product_dependency="COMPARE_WITH_SAVED_LOCAL100_AND_BIND_OPT_IN_PROVIDER",
        )
    except Exception as error:
        result.update(
            status="DEVELOPMENT_LOCAL100_INPUT_BLOCKED",
            blocker={"stage": stage, "type": type(error).__name__, "detail": str(error)},
        )

    raw = json.dumps(
        result, ensure_ascii=False, sort_keys=True, indent=2, allow_nan=False
    ).encode("utf-8")
    digest = sha(raw)
    relative = reports / f"C120.R11_development_local100_input.{digest}.json"
    path = root / relative
    try:
        with path.open("xb") as handle:
            handle.write(raw)
            handle.flush()
            os.fsync(handle.fileno())
    except FileExistsError:
        pass
    require(path.read_bytes() == raw, "C120_REPORT_RELOAD_CONFLICT")
    result["output"] = {
        "path": relative.as_posix(),
        "file_bytes_sha256": digest,
        "reload_validation": "PASS",
    }
    detail_fields = {
        "hybrid100_order", "scored_pool50_order", "local100_order",
        "scores_by_chunk_id", "chunk_text_sha256_by_id",
    }
    print(json.dumps(
        {key: value for key, value in result.items() if key not in detail_fields},
        ensure_ascii=False, indent=2,
    ))
    return result


V3_C120 = run_v3_c120()
````

### سل ۱۲۱ — مقایسهٔ Local100 بازسازی‌شده با خروجی اصلی NB22B

برای همان سؤال Development، ترتیب کامل C120 با ردیف `local_onnx_depth100`
در `A05.development_variants` مقایسه می‌شود. هش هر دو فایل، یکتایی ردیف،

امتیازدهی یا بازسازی تازه‌ای اجرا نمی‌کند.

</div>

````python
# C121 — R11.3 — Compare one Development Local100 order with NB22B A05
def run_v3_c121():
    def require(ok, code):
        if not ok:
            raise ValueError(code)

    def sha(raw):
        return hashlib.sha256(raw).hexdigest()

    reports = Path("reports/phase2/quality_improvement/NB22E.lab.v3")
    root = next(
        (p for p in (Path.cwd(), *Path.cwd().parents) if (p / reports).is_dir()),
        None,
    )
    require(root is not None, "C121_PROJECT_ROOT_NOT_FOUND")

    c120_sha = "33578338372385982a433939fbec3e140cac3495d0ef8228847cceb99ac05975"
    c120_path = reports / f"C120.R11_development_local100_input.{c120_sha}.json"
    raw = (root / c120_path).read_bytes()
    require(sha(raw) == c120_sha, "C121_C120_HASH_CONFLICT")
    rebuilt = json.loads(raw)
    require(rebuilt["status"] == "ONE_DEVELOPMENT_LOCAL100_INPUT_BUILT",
            "C121_SUCCESSFUL_C120_REQUIRED")

    candidate = rebuilt["candidate_id"]
    actual = rebuilt["local100_order"]
    hybrid = rebuilt["hybrid100_order"]
    require(len(actual) == len(hybrid) == len(set(actual)) == 100
            and set(actual) == set(hybrid), "C121_C120_POOL_CONFLICT")

    result = {
        "cell": "C121",
        "roadmap_action": "R11",
        "roadmap_subtask": "R11.3",
        "candidate_id": candidate,
        "input_file_hashes": {"C120": c120_sha},
        "status": "SAVED_LOCAL100_COMPARISON_BLOCKED",
        "local100_matches_saved_order_verified": False,
        "product_runtime_binding_verified": False,
        "product_regression_tests_executed": 0,
        "new_api_calls": 0,
        "new_local_pairs_scored": 0,
        "rank_reconstruction_for_unverified_candidates": "DEFERRED",
        "done_item_count": 7,
        "open_item_count": 5,
        "newly_completed_items_from_previous_open_set": [],
        "historical_gate": "FAIL",
        "selected_version_gate": "BLOCKED",
    }

    source_dir = Path("reports/phase2/quality_improvement/NB22B.v1")
    stage = "LOCATE_NB22B_A05"
    try:
        paths = sorted((root / source_dir).glob("A05.development_variants.*.jsonl"))
        result["discovered_source_paths"] = [
            p.relative_to(root).as_posix() for p in paths
        ]
        require(bool(paths), "C121_NB22B_A05_FILE_NOT_FOUND")

        stage = "VERIFY_NB22B_A05_AND_BIND_ROW"
        matches = []
        for path in paths:
            expected = path.name.removeprefix("A05.development_variants.").removesuffix(".jsonl")
            require(len(expected) == 64 and all(c in "0123456789abcdef" for c in expected),
                    f"C121_INVALID_CONTENT_ADDRESS:{path.name}")
            source = path.read_bytes()
            require(sha(source) == expected, f"C121_A05_HASH_CONFLICT:{path.name}")
            rows = []
            for line_number, line in enumerate(source.splitlines(), start=1):
                if not line.strip():
                    continue
                row = json.loads(line)
                if (row["candidate_id"], row["variant"], row["split"]) == (
                    candidate, "local_onnx_depth100", "development"
                ):
                    rows.append((line_number, row["ranked_chunk_ids"]))
            require(len(rows) <= 1, f"C121_DUPLICATE_REFERENCE_ROW:{path.name}")
            if rows:
                line_number, order = rows[0]
                matches.append({
                    "path": path.relative_to(root).as_posix(),
                    "file_bytes_sha256": expected,
                    "line_number": line_number,
                    "order": order,
                })

        require(bool(matches), "C121_LOCAL100_ROW_NOT_FOUND_IN_A05")
        result["reference_receipts"] = [
            {k: v for k, v in match.items() if k != "order"} for match in matches
        ]
        saved = matches[0]["order"]
        require(all(match["order"] == saved for match in matches),
                "C121_CONFLICTING_SAVED_LOCAL100_ORDERS")

        stage = "COMPARE_COMPLETE_ORDER"
        require(len(saved) == len(set(saved)) == 100, "C121_SAVED_ORDER_NOT_DEPTH100")
        result["same_chunk_set"] = set(saved) == set(actual)
        result["only_in_saved"] = sorted(set(saved) - set(actual))
        result["only_in_c120"] = sorted(set(actual) - set(saved))
        differences = [
            {"rank": rank, "saved": old, "c120": new}
            for rank, (old, new) in enumerate(zip(saved, actual), start=1)
            if old != new
        ]
        matched = not differences
        result.update(
            status="SAVED_LOCAL100_ORDER_MATCHED" if matched else "SAVED_LOCAL100_ORDER_DIFFERS",
            local100_matches_saved_order_verified=matched,
            different_position_count=len(differences),
            first_difference=differences[0] if differences else None,
            blocker=None,
            next_product_dependency=(
                "VERIFY_QUOTA100_AND_CHAR_RRF_CHAIN_AT_OPT_IN_ENTRYPOINT"
                if matched else "EXPLAIN_FIRST_LOCAL100_ORDER_DIFFERENCE"
            ),
        )
    except (OSError, ValueError, KeyError, TypeError) as error:
        result["blocker"] = {"stage": stage, "detail": str(error)}

    payload = json.dumps(result, ensure_ascii=False, sort_keys=True, indent=2).encode("utf-8")
    digest = sha(payload)
    relative = reports / f"C121.R11_saved_local100_order_comparison.{digest}.json"
    target = root / relative
    try:
        with target.open("xb") as handle:
            handle.write(payload)
            handle.flush()
            os.fsync(handle.fileno())
    except FileExistsError:
        pass
    require(target.read_bytes() == payload, "C121_REPORT_RELOAD_CONFLICT")
    result["output"] = {
        "path": relative.as_posix(),
        "file_bytes_sha256": digest,
        "reload_validation": "PASS",
    }
    print(json.dumps(result, ensure_ascii=False, indent=2))
    return result


V3_C121 = run_v3_c121()
````

### سل ۱۲۲ — تطبیق دو مرحلهٔ رتبه‌بندی برای یک سؤال Development

رسیدهای سل‌های ۱۲۰ و ۱۲۱ مصرف می‌شوند. ترتیب Quota100 با خروجی ذخیره‌شدهٔ NB22C مقایسه می‌شود. پس از تطابق کامل، کانال char ذخیره‌شده به همان ترتیب اعمال و خروجی کامل char_rrf با NB22D مقایسه می‌شود.

سل آفلاین است و رتبه‌های ۱۶ پروندهٔ معوق را بازسازی نمی‌کند. نتیجه در دامنهٔ همان یک سؤال Development ثبت می‌شود.

</div>

````python
# C122 — R11.3 — Verify Quota100 and char_rrf against saved Development orders
def run_v3_c122():
    def require(ok, message):
        if not ok:
            raise ValueError(message)

    def sha(raw):
        return hashlib.sha256(raw).hexdigest()

    base = Path("reports/phase2/quality_improvement")
    reports = base / "NB22E.lab.v3"
    root = next(
        (p for p in (Path.cwd(), *Path.cwd().parents) if (p / reports).is_dir()),
        None,
    )
    require(root is not None, "C122_PROJECT_ROOT_NOT_FOUND")

    def read_bound(path, digest):
        raw = (root / path).read_bytes()
        require(sha(raw) == digest, f"C122_HASH_CONFLICT:{path}")
        return raw

    c120_sha = "33578338372385982a433939fbec3e140cac3495d0ef8228847cceb99ac05975"
    c121_sha = "c1123b6101d85d5564ec82570e6e36b51cf0c6263bf7836eff71187c5446e3fe"
    inputs = json.loads(read_bound(
        reports / f"C120.R11_development_local100_input.{c120_sha}.json",
        c120_sha,
    ))
    parity = json.loads(read_bound(
        reports / f"C121.R11_saved_local100_order_comparison.{c121_sha}.json",
        c121_sha,
    ))
    candidate = inputs["candidate_id"]
    require(
        inputs["status"] == "ONE_DEVELOPMENT_LOCAL100_INPUT_BUILT"
        and parity["candidate_id"] == candidate
        and parity["input_file_hashes"]["C120"] == c120_sha
        and parity["local100_matches_saved_order_verified"] is True,
        "C122_C120_C121_BINDING_CONFLICT",
    )

    result = {
        "cell": "C122",
        "roadmap_action": "R11",
        "roadmap_subtask": "R11.3",
        "scope": "ONE_DEVELOPMENT_QUERY_WITH_SAVED_CHAR_CHANNEL",
        "candidate_id": candidate,
        "question_sha256": inputs["question_sha256"],
        "input_file_hashes": {"C120": c120_sha, "C121": c121_sha},
        "comparisons": {},
        "product_runtime_binding_verified": False,
        "product_regression_tests_executed": 0,
        "new_api_calls": 0,
        "new_local_pairs_scored": 0,
        "target_case_rankings_reconstructed": 0,
        "rank_reconstruction_for_unverified_candidates": "DEFERRED",
        "done_item_count": 7,
        "open_item_count": 5,
        "newly_completed_items_from_previous_open_set": [],
        "historical_gate": "FAIL",
        "selected_version_gate": "BLOCKED",
    }

    def read_row(role, directory, stem, digest, variant=None):
        path = base / directory / f"{stem}.{digest}.jsonl"
        raw = read_bound(path, digest)
        matches = []
        for line_number, line in enumerate(raw.splitlines(), start=1):
            if not line.strip():
                continue
            row = json.loads(line)
            if row["candidate_id"] != candidate:
                continue
            if variant is not None and (
                row["variant"] != variant or row["split"] != "development"
            ):
                continue
            matches.append((line_number, row))
        require(len(matches) == 1, f"C122_EXPECTED_ONE_ROW:{role}:{len(matches)}")
        line_number, row = matches[0]
        result["input_file_hashes"][role] = digest
        result.setdefault("source_rows", {})[role] = {
            "path": path.as_posix(),
            "line_number": line_number,
        }
        return row

    def compare(label, observed, saved):
        first_difference = next(
            (
                {"rank": i + 1, "observed": current, "saved": previous}
                for i, (current, previous) in enumerate(zip(observed, saved))
                if current != previous
            ),
            None,
        )
        if first_difference is None and len(observed) != len(saved):
            index = min(len(observed), len(saved))
            first_difference = {
                "rank": index + 1,
                "observed": observed[index] if index < len(observed) else None,
                "saved": saved[index] if index < len(saved) else None,
            }
        result["comparisons"][label] = {
            "full_order_matches": observed == saved,
            "observed_depth": len(observed),
            "saved_depth": len(saved),
            "first_difference": first_difference,
        }
        require(observed == saved, f"C122_FULL_ORDER_DIFFERS:{label}")

    stage = "QUOTA100"
    try:
        hybrid = inputs["hybrid100_order"]
        local = inputs["local100_order"]
        pool = inputs["scored_pool50_order"]
        require(
            len(hybrid) == len(local) == len(set(hybrid)) == 100
            and set(hybrid) == set(local),
            "C122_DEPTH100_POOL_CONFLICT",
        )
        old_pool = set(pool)
        require(len(pool) == len(old_pool) == 50 and old_pool == set(hybrid[:50]),
                "C122_POOL50_MEMBERSHIP_CONFLICT")

        # NB22C quota100_k10_w0.75_n1 uses Pool50 membership.
        hybrid_rank = {cid: i for i, cid in enumerate(hybrid, start=1)}
        local_rank = {cid: i for i, cid in enumerate(local, start=1)}
        ranked = sorted(hybrid, key=lambda cid: (
            -(0.75 / (10 + local_rank[cid]) + 0.25 / (10 + hybrid_rank[cid])),
            cid,
        ))
        top = []
        newcomers = 0
        for cid in ranked:
            if cid not in old_pool and newcomers >= 1:
                continue
            top.append(cid)
            newcomers += int(cid not in old_pool)
            if len(top) == 10:
                break
        selected = set(top)
        quota_order = top + [cid for cid in ranked if cid not in selected]

        saved_quota = read_row(
            "NB22C_A05", "NB22C.v1", "A05.retained_development_rankings",
            "5bc216aae839bc80cc2f184ca3a3cd3e4ac33657655c3bffb4fac410e96e9190",
            "quota100_k10_w0.75_n1",
        )
        compare("quota100", quota_order, saved_quota["ranked_chunk_ids"])

        stage = "CHAR_RRF"
        channel = read_row(
            "NB22D_A03", "NB22D.v1", "A03.lexical_channels",
            "41d1e2481910dee526c066ee807bc27ce22486d841b150bacace4c8c8ab0aa82",
        )["channels"]["char"]
        require(len(channel) == len(set(channel)), "C122_DUPLICATE_CHAR_CHANNEL_ID")

        # NB22D char_rrf combines the quota order and the saved char order.
        scores = {}
        for order, weight in ((quota_order, 0.75), (channel, 0.25)):
            for rank, cid in enumerate(order, start=1):
                scores[cid] = scores.get(cid, 0.0) + weight / (10 + rank)
        fused = sorted(scores, key=lambda cid: (-scores[cid], cid))

        saved_fused = read_row(
            "NB22D_A05", "NB22D.v1", "A05.retained_development_rankings",
            "ab25d43485bd940f964cf872606f52c24809f52b4acbb7d206046ae01f03f633",
            "char_rrf",
        )
        compare("char_rrf", fused, saved_fused["ranked_chunk_ids"])
        result.update(
            status="QUOTA100_AND_CHAR_RRF_FULL_ORDERS_MATCHED",
            blocker=None,
            quota100_order=quota_order,
            char_rrf_order=fused,
            char_rrf_top10=fused[:10],
            next_product_dependency="EXERCISE_OPT_IN_ENTRYPOINT_WITH_VERIFIED_ORDER_AND_RECEIPTS",
        )
    except (OSError, ValueError, KeyError, TypeError) as error:
        result.update(
            status="DEVELOPMENT_RANK_CHAIN_BLOCKED",
            blocker={"stage": stage, "type": type(error).__name__, "detail": str(error)},
        )

    raw = json.dumps(result, ensure_ascii=False, sort_keys=True, indent=2).encode("utf-8")
    digest = sha(raw)
    relative = reports / f"C122.R11_development_rank_chain.{digest}.json"
    target = root / relative
    try:
        with target.open("xb") as handle:
            handle.write(raw)
            handle.flush()
            os.fsync(handle.fileno())
    except FileExistsError:
        pass
    require(target.read_bytes() == raw, "C122_REPORT_RELOAD_CONFLICT")
    result["output"] = {
        "path": relative.as_posix(),
        "file_bytes_sha256": digest,
        "reload_validation": "PASS",
    }
    print(json.dumps(
        {k: v for k, v in result.items() if k not in {"quota100_order", "char_rrf_order"}},
        ensure_ascii=False,
        indent=2,
    ))
    return result


V3_C122 = run_v3_c122()
````

### سل ۱۲۳ — تطبیق رسید اجرای اختیاری با ترتیب تأییدشده

رسیدهای ورودی و خروجی سل ۱۱۵ با هش خوانده می‌شوند. هویت سؤال و ترتیب دقیق ده قطعه با سل ۱۲۲ تطبیق داده می‌شود. ثبت رسید ورودی پیش از فراخوانی مولد آزمایشی و اتصال هش رسید ورودی به رسید خروجی کنترل می‌شود.

دامنه یک سؤال Development و یک اجرای آفلاین موجود است. امتیازدهی، اجرای دوبارهٔ مولد و فراخوانی API در این سل انجام نمی‌شود.

````python
# C123 — R11.3 — Bind C122 rank-chain evidence to existing C115 replay receipts
def run_v3_c123():
    def require(ok, code):
        if not ok:
            raise ValueError(code)

    def sha(raw):
        return hashlib.sha256(raw).hexdigest()

    reports = Path("reports/phase2/quality_improvement/NB22E.lab.v3")
    root = next(
        (p for p in (Path.cwd(), *Path.cwd().parents) if (p / reports).is_dir()),
        None,
    )
    require(root is not None, "C123_PROJECT_ROOT_NOT_FOUND")
    root = root.resolve()

    def read_bound(relative, digest):
        path = (root / relative).resolve()
        require(path.is_relative_to(root), "C123_PATH_OUTSIDE_PROJECT")
        raw = path.read_bytes()
        require(sha(raw) == digest, f"C123_HASH_CONFLICT:{relative}")
        return json.loads(raw)

    def pinned(stem, digest):
        return read_bound(reports / f"{stem}.{digest}.json", digest)

    c115_sha = "42aed73b66bca6bcd6443383ba97314f2ccab7d35d415f1bdfc21e4cef970cf9"
    c122_sha = "85492d51e95b2aa7dcb2a1a5f5cdb44a0bd5635550d736e38bb20dd5af8d209e"
    result = {
        "cell": "C123",
        "roadmap_action": "R11",
        "roadmap_subtask": "R11.3",
        "scope": "ONE_DEVELOPMENT_QUERY_EXISTING_OPT_IN_REPLAY_WITH_STUB",
        "input_file_hashes": {"C115": c115_sha, "C122": c122_sha},
        "offline_replay_binding_verified": False,
        "product_runtime_binding_verified": False,
        "product_regression_tests_executed": 0,
        "new_api_calls": 0,
        "new_local_pairs_scored": 0,
        "target_case_rankings_reconstructed": 0,
        "rank_reconstruction_for_unverified_candidates": "DEFERRED",
        "done_item_count": 7,
        "open_item_count": 5,
        "newly_completed_items_from_previous_open_set": [],
        "historical_gate": "FAIL",
        "selected_version_gate": "BLOCKED",
    }

    stage = "BIND_REPLAY_AND_RANK_CHAIN"
    try:
        replay = pinned("C115.R11_development_rank_replay", c115_sha)
        chain = pinned("C122.R11_development_rank_chain", c122_sha)
        require(
            replay["status"] == "ONE_DEVELOPMENT_RANKING_REPLAY_AND_RECEIPT_CHAIN_PASSED"
            and replay["scope"] == "STORED_DEVELOPMENT_INPUTS_WITH_STUB_GENERATOR"
            and replay["full_order_matches_saved_nb22d"] is True,
            "C123_C115_REPLAY_NOT_ESTABLISHED",
        )
        require(
            chain["status"] == "QUOTA100_AND_CHAR_RRF_FULL_ORDERS_MATCHED"
            and all(chain["comparisons"][key]["full_order_matches"] is True
                    for key in ("quota100", "char_rrf")),
            "C123_C122_ORDER_PARITY_REQUIRED",
        )
        top10 = chain["char_rrf_order"][:10]
        require(
            len(set(top10)) == 10
            and top10 == chain["char_rrf_top10"] == replay["top10"],
            "C123_TOP10_ORDER_CONFLICT",
        )
        require(
            replay["candidate_id"] == chain["candidate_id"]
            and replay["question_sha256"] == chain["question_sha256"],
            "C123_QUERY_IDENTITY_CONFLICT",
        )
        result.update(
            candidate_id=chain["candidate_id"],
            question_sha256=chain["question_sha256"],
            ordered_chunk_ids=top10,
        )

        stage = "VERIFY_REPLAY_RECEIPTS"
        verified_receipts = {}
        for role in ("input", "output"):
            descriptor = replay["receipts"][role]
            verified_receipts[role] = read_bound(
                descriptor["path"], descriptor["file_bytes_sha256"]
            )
            result["input_file_hashes"][f"C115_{role}_receipt"] = (
                descriptor["file_bytes_sha256"]
            )

        receipt = verified_receipts["input"]
        require(
            receipt.get("candidate_id") == chain["candidate_id"]
            and receipt.get("question_sha256") == chain["question_sha256"]
            and receipt.get("selected_variant") == "char_rrf"
            and receipt.get("ordered_chunk_ids") == top10,
            "C123_INPUT_RECEIPT_BINDING_CONFLICT",
        )
        require(
            replay["events"] == [
                "input_receipt_written",
                "stub_generator_verified_input",
                "output_receipt_written",
            ],
            "C123_RECEIPT_EVENT_ORDER_CONFLICT",
        )

        stage = "VERIFY_OUTPUT_TO_INPUT_LINK"
        output_receipt = verified_receipts["output"]
        expected_link = result["input_file_hashes"]["C115_input_receipt"]
        link = output_receipt.get("input_receipt_file_bytes_sha256")
        if link is None:
            nested = output_receipt.get("input_receipt")
            if isinstance(nested, dict):
                link = nested.get("file_bytes_sha256")
        result["output_receipt_fields"] = sorted(output_receipt)
        require(link == expected_link, "C123_OUTPUT_INPUT_HASH_LINK_NOT_VERIFIED")

        result.update(
            status="VERIFIED_RANK_CHAIN_BOUND_TO_EXISTING_OPT_IN_REPLAY",
            offline_replay_binding_verified=True,
            blocker=None,
            next_product_dependency="BIND_REAL_VERSIONED_PROVIDERS_AT_OPT_IN_ENTRYPOINT",
        )
    except (OSError, ValueError, KeyError, TypeError) as error:
        result.update(
            status="EXISTING_OPT_IN_REPLAY_BINDING_BLOCKED",
            blocker={"stage": stage, "type": type(error).__name__, "detail": str(error)},
        )

    raw = json.dumps(result, ensure_ascii=False, sort_keys=True, indent=2).encode("utf-8")
    digest = sha(raw)
    relative = reports / f"C123.R11_verified_rank_chain_replay_binding.{digest}.json"
    target = root / relative
    try:
        with target.open("xb") as handle:
            handle.write(raw)
            handle.flush()
            os.fsync(handle.fileno())
    except FileExistsError:
        pass
    require(target.read_bytes() == raw, "C123_REPORT_RELOAD_CONFLICT")
    result["output"] = {
        "path": relative.as_posix(),
        "file_bytes_sha256": digest,
        "reload_validation": "PASS",
    }
    print(json.dumps(result, ensure_ascii=False, indent=2))
    return result


V3_C123 = run_v3_c123()
````

### سل ۱۲۴ — اجرای مسیر اختیاری با فراهم‌کنندهٔ نسخه‌بندی‌شده

تابع `ask_with_char_rrf` از سورس احرازشدهٔ محصول اجرا می‌شود. فراهم‌کننده از Hybrid100، Local100، عضویت Pool50 و کانال char ثبت‌شده، ترتیب را در زمان فراخوانی می‌سازد و آن را با خروجی کامل سل ۱۲۲ تطبیق می‌دهد.

مولد آزمایشی وجود و صحت رسید ورودی را پیش از بازگرداندن پاسخ کنترل می‌کند. سپس هش پاسخ و اتصال رسید خروجی به رسید ورودی بررسی می‌شود. دامنه یک سؤال Development است. اجرای سل آفلاین است و رتبه‌های ۱۶ پروندهٔ معوق بازسازی نمی‌شوند.

</div>

````python
# C124 — R11.3 — Version-bound provider through the product opt-in method
def run_v3_c124():
    def require(ok, code):
        if not ok:
            raise ValueError(code)

    def sha(raw):
        return hashlib.sha256(raw).hexdigest()

    def text_sha(value):
        return sha(value.encode("utf-8"))

    reports = Path("reports/phase2/quality_improvement/NB22E.lab.v3")
    root = next(
        (p for p in (Path.cwd(), *Path.cwd().parents) if (p / reports).is_dir()),
        None,
    )
    require(root is not None, "C124_PROJECT_ROOT_NOT_FOUND")
    root = root.resolve()

    def read_bound(relative, digest):
        path = (root / relative).resolve()
        require(path.is_relative_to(root), "C124_PATH_OUTSIDE_PROJECT")
        raw = path.read_bytes()
        require(sha(raw) == digest, f"C124_HASH_CONFLICT:{relative}")
        return raw

    def read_report(stem, digest):
        return json.loads(read_bound(reports / f"{stem}.{digest}.json", digest))

    def save(stem, value):
        raw = json.dumps(
            value, ensure_ascii=False, sort_keys=True, indent=2, allow_nan=False
        ).encode("utf-8")
        digest = sha(raw)
        relative = reports / f"{stem}.{digest}.json"
        path = root / relative
        try:
            with path.open("xb") as handle:
                handle.write(raw)
                handle.flush()
                os.fsync(handle.fileno())
        except FileExistsError:
            pass
        require(path.read_bytes() == raw, "C124_RECEIPT_RELOAD_CONFLICT")
        return {"path": relative.as_posix(), "file_bytes_sha256": digest,
                "reload_validation": "PASS"}

    c120_sha = "33578338372385982a433939fbec3e140cac3495d0ef8228847cceb99ac05975"
    c122_sha = "85492d51e95b2aa7dcb2a1a5f5cdb44a0bd5635550d736e38bb20dd5af8d209e"
    c123_sha = "7cd12056267bc9ed088f3d36eefcb48f945159af4c87da60ddea4a65d5e2d656"
    pipeline_sha = "99ee048d00a7ad1884540fa297eb41df86ad4e54f6c49f26385ebb6b2ed504f6"
    pipeline_path = "src/pipeline/rag_pipeline.py"
    result = {
        "cell": "C124",
        "roadmap_action": "R11",
        "roadmap_subtask": "R11.3",
        "scope": "ONE_DEVELOPMENT_QUERY_STORED_INPUT_PROVIDERS_STUB_GENERATOR",
        "input_file_hashes": {
            "C120": c120_sha, "C122": c122_sha, "C123": c123_sha,
            "pipeline_source": pipeline_sha,
        },
        "stored_input_opt_in_integration_verified": False,
        "product_runtime_binding_verified": False,
        "product_regression_tests_executed": 0,
        "new_api_calls": 0,
        "new_local_pairs_scored": 0,
        "target_case_rankings_reconstructed": 0,
        "rank_reconstruction_for_unverified_candidates": "DEFERRED",
        "done_item_count": 7,
        "open_item_count": 5,
        "newly_completed_items_from_previous_open_set": [],
        "historical_gate": "FAIL",
        "selected_version_gate": "BLOCKED",
    }
    events = []
    receipts = {}

    stage = "BIND_PRIOR_EVIDENCE"
    try:
        local = read_report("C120.R11_development_local100_input", c120_sha)
        chain = read_report("C122.R11_development_rank_chain", c122_sha)
        replay = read_report("C123.R11_verified_rank_chain_replay_binding", c123_sha)
        candidate = chain["candidate_id"]
        question_hash = chain["question_sha256"]
        require(
            local["status"] == "ONE_DEVELOPMENT_LOCAL100_INPUT_BUILT"
            and chain["status"] == "QUOTA100_AND_CHAR_RRF_FULL_ORDERS_MATCHED"
            and replay["offline_replay_binding_verified"] is True
            and chain["input_file_hashes"]["C120"] == c120_sha
            and replay["input_file_hashes"]["C122"] == c122_sha
            and local["candidate_id"] == replay["candidate_id"] == candidate
            and local["question_sha256"] == replay["question_sha256"] == question_hash
            and replay["ordered_chunk_ids"] == chain["char_rrf_top10"],
            "C124_PRIOR_EVIDENCE_CONFLICT",
        )
        result.update(candidate_id=candidate, question_sha256=question_hash)

        stage = "BIND_QUESTION_CHUNKS_AND_CHAR_CHANNEL"
        pins = local["input_file_hashes"]
        c056_sha = pins["NB22_C056"]
        c056_path = (
            Path("reports/phase2/release/NB22.C056.retrieval_verification")
            / f"C056.independent_retrieval_metric_verification.{c056_sha}.json"
        )
        c056 = json.loads(read_bound(c056_path, c056_sha))

        def load_selected(role, key, wanted):
            binding = c056["input_file_receipts"][role]
            digest = binding["file_bytes_sha256"]
            require(digest == pins[role], f"C124_INPUT_PIN_CONFLICT:{role}")
            raw = read_bound(binding["path"], digest)
            selected = {}
            for line in raw.splitlines():
                if line.strip():
                    row = json.loads(line)
                    identity = row[key]
                    if identity in wanted:
                        require(identity not in selected, f"C124_DUPLICATE_ID:{role}")
                        selected[identity] = row
            require(set(selected) == wanted, f"C124_MISSING_ID:{role}")
            result["input_file_hashes"][role] = digest
            return selected

        golden = load_selected("golden", "candidate_id", {candidate})
        question = golden[candidate]["evaluation_content"]["question"].strip()
        require(text_sha(question) == question_hash, "C124_QUESTION_HASH_CONFLICT")
        expected_top10 = chain["char_rrf_top10"]
        corpus = load_selected("chunks", "chunk_id", set(expected_top10))

        channel_ref = chain["source_rows"]["NB22D_A03"]
        channel_sha = chain["input_file_hashes"]["NB22D_A03"]
        channel_raw = read_bound(channel_ref["path"], channel_sha)
        channel_row = json.loads(channel_raw.splitlines()[channel_ref["line_number"] - 1])
        require(channel_row["candidate_id"] == candidate, "C124_CHAR_CHANNEL_ID_CONFLICT")
        channel = channel_row["channels"]["char"]
        require(len(channel) == len(set(channel)), "C124_DUPLICATE_CHAR_ID")
        result["input_file_hashes"]["NB22D_A03"] = channel_sha

        stage = "LOAD_PINNED_PRODUCT_METHODS"
        source = read_bound(pipeline_path, pipeline_sha).decode("utf-8")
        tree = ast.parse(source, filename=pipeline_path)
        names = {"ask_with_char_rrf", "build_context", "format_sources"}
        classes = [
            node for node in tree.body if isinstance(node, ast.ClassDef)
            and names <= {item.name for item in node.body if isinstance(item, ast.FunctionDef)}
        ]
        require(len(classes) == 1, "C124_PRODUCT_METHOD_CONTAINER_CONFLICT")
        methods = [
            ast.get_source_segment(source, item)
            for item in classes[0].body
            if isinstance(item, ast.FunctionDef) and item.name in names
        ]
        # Load only the pinned methods. Product constructors and clients stay idle.
        method_scope = {}
        class_source = "class C124Probe:\n" + "\n\n".join(
            "    " + method for method in methods
        )
        exec(compile(class_source, pipeline_path, "exec"), method_scope)
        probe = method_scope["C124Probe"]()

        def rank_sources(*, question, candidate_id, top_k):
            require(
                candidate_id == candidate and text_sha(question) == question_hash
                and top_k == 10,
                "C124_PROVIDER_QUERY_CONFLICT",
            )
            hybrid = local["hybrid100_order"]
            local_order = local["local100_order"]
            pool = set(local["scored_pool50_order"])
            require(
                len(hybrid) == len(local_order) == len(set(hybrid)) == 100
                and set(hybrid) == set(local_order)
                and len(pool) == 50 and pool == set(hybrid[:50]),
                "C124_PROVIDER_POOL_CONFLICT",
            )
            hr = {cid: rank for rank, cid in enumerate(hybrid, 1)}
            lr = {cid: rank for rank, cid in enumerate(local_order, 1)}
            ranked = sorted(
                hybrid, key=lambda cid: (
                    -(0.75 / (10 + lr[cid]) + 0.25 / (10 + hr[cid])), cid
                ),
            )
            top = []
            newcomers = 0
            for cid in ranked:
                if cid not in pool and newcomers == 1:
                    continue
                top.append(cid)
                newcomers += int(cid not in pool)
                if len(top) == 10:
                    break
            chosen = set(top)
            quota = top + [cid for cid in ranked if cid not in chosen]
            require(quota == chain["quota100_order"], "C124_QUOTA100_ORDER_CONFLICT")

            scores = {}
            for order, weight in ((quota, 0.75), (channel, 0.25)):
                for rank, cid in enumerate(order, 1):
                    scores[cid] = scores.get(cid, 0.0) + weight / (10 + rank)
            fused = sorted(scores, key=lambda cid: (-scores[cid], cid))
            require(
                fused == chain["char_rrf_order"] and fused[:top_k] == expected_top10,
                "C124_CHAR_RRF_ORDER_CONFLICT",
            )
            events.append("provider_verified_full_orders")
            return {
                "candidate_id": candidate,
                "question_sha256": question_hash,
                "selected_variant": "char_rrf",
                "runtime_identity": {
                    "scope": "HASH_BOUND_STORED_DEVELOPMENT_INPUTS",
                    "pipeline_source_sha256": pipeline_sha,
                    "local100_receipt_sha256": c120_sha,
                    "rank_chain_receipt_sha256": c122_sha,
                    "char_channel_file_sha256": channel_sha,
                },
                "ordered_chunks": [
                    {"chunk_id": cid, "text": corpus[cid]["content"],
                     "title": corpus[cid].get("title"), "url": corpus[cid].get("url")}
                    for cid in fused[:top_k]
                ],
            }

        def write_receipt(role, value):
            require(role in {"input", "output"} and role not in receipts,
                    "C124_RECEIPT_ROLE_CONFLICT")
            receipt = save(f"C124.R11_opt_in_{role}_receipt", value)
            receipts[role] = receipt
            events.append(f"{role}_receipt_written")
            return receipt["file_bytes_sha256"]

        answer = "در منابع موجود پاسخ کافی پیدا نشد."

        class StubGenerator:
            def generate_answer(self, *, question, context):
                require("input" in receipts and "output" not in receipts,
                        "C124_PRE_GENERATION_RECEIPT_REQUIRED")
                bound = receipts["input"]
                recorded = json.loads(read_bound(bound["path"], bound["file_bytes_sha256"]))
                require(
                    recorded["candidate_id"] == candidate
                    and recorded["question_sha256"] == text_sha(question) == question_hash
                    and recorded["ordered_chunk_ids"] == expected_top10
                    and recorded["ordered_chunk_text_sha256"] == [
                        text_sha(corpus[cid]["content"]) for cid in expected_top10
                    ]
                    and recorded["rendered_context_sha256"] == text_sha(context),
                    "C124_GENERATOR_INPUT_RECEIPT_CONFLICT",
                )
                events.append("stub_verified_persisted_input")
                return answer

        stage = "RUN_OPT_IN_AND_VERIFY_RECEIPTS"
        probe.generator = StubGenerator()
        response = probe.ask_with_char_rrf(
            question, candidate_id=candidate, rank_sources=rank_sources,
            write_receipt=write_receipt, top_k=10,
        )
        require(events == [
            "provider_verified_full_orders", "input_receipt_written",
            "stub_verified_persisted_input", "output_receipt_written",
        ], "C124_EVENT_ORDER_CONFLICT")

        input_sha = receipts["input"]["file_bytes_sha256"]
        output = receipts["output"]
        output_payload = json.loads(read_bound(output["path"], output["file_bytes_sha256"]))
        require(
            response["answer"] == answer and response["sources"] == []
            and response["selected_variant"] == "char_rrf"
            and response["from_cache"] is False
            and response["input_receipt_file_bytes_sha256"] == input_sha
            and response["output_receipt_file_bytes_sha256"] == output["file_bytes_sha256"]
            and output_payload["input_receipt_file_bytes_sha256"] == input_sha
            and output_payload["answer_text_utf8_sha256"] == text_sha(response["answer"]),
            "C124_OUTPUT_RECEIPT_CONFLICT",
        )
        result.update(
            status="VERSION_BOUND_STORED_INPUT_PROVIDER_PASSED_OPT_IN_PROBE",
            stored_input_opt_in_integration_verified=True,
            blocker=None,
            next_product_dependency="BIND_LIVE_VERSIONED_PROVIDERS_AT_PRODUCT_ENTRYPOINT",
        )
    except (OSError, ValueError, KeyError, TypeError, IndexError, SyntaxError) as error:
        result.update(
            status="VERSION_BOUND_OPT_IN_PROBE_BLOCKED",
            blocker={"stage": stage, "type": type(error).__name__, "detail": str(error)},
        )

    result.update(events=events, receipts=receipts)
    result["output"] = save("C124.R11_version_bound_provider_probe", result)
    print(json.dumps(result, ensure_ascii=False, indent=2))
    return result


V3_C124 = run_v3_c124()
````

````python
# C125 — R11.3 — Check NB16 Hybrid100 on one Development query
def run_v3_c125():
    def require(ok, code):
        if not ok:
            raise ValueError(code)

    def sha(raw):
        return hashlib.sha256(raw).hexdigest()

    reports = Path("reports/phase2/quality_improvement/NB22E.lab.v3")
    root = next(
        (p for p in (Path.cwd(), *Path.cwd().parents) if (p / reports).is_dir()),
        None,
    )
    require(root is not None, "C125_PROJECT_ROOT_NOT_FOUND")
    root = root.resolve()

    def read_bound(relative, digest):
        path = (root / relative).resolve()
        require(path.is_relative_to(root), "C125_PATH_OUTSIDE_PROJECT")
        raw = path.read_bytes()
        require(sha(raw) == digest, f"C125_HASH_CONFLICT:{relative}")
        return raw

    nb_path = "notebooks/phase2/16_hybrid_retrieval.ipynb"
    nb_sha = "40699ca54c57fcef40472df6b6fc42a54f74dc2e9bdc9bb40e0ccc93bb7ad6a7"
    c120_sha = "33578338372385982a433939fbec3e140cac3495d0ef8228847cceb99ac05975"
    result = {
        "cell": "C125",
        "roadmap_subtask": "R11.3",
        "input_file_hashes": {"NB16": nb_sha, "C120": c120_sha},
        "full_hybrid100_order_matches": False,
        "product_runtime_binding_verified": False,
        "done_item_count": 7,
        "open_item_count": 5,
        "newly_completed_items_from_previous_open_set": [],
        "selected_version_gate": "BLOCKED",
        "target_case_rankings_reconstructed": 0,
        "embedding_calls": 0,
        "generation_calls": 0,
        "qdrant_queries_attempted": 0,
    }

    stage = "VERIFY_SOURCE_AND_REFERENCE"
    try:
        nb = json.loads(read_bound(nb_path, nb_sha))
        source = "".join(nb["cells"][37]["source"])
        tree = ast.parse(source, filename=nb_path)
        prior = json.loads(read_bound(
            reports / f"C120.R11_development_local100_input.{c120_sha}.json",
            c120_sha,
        ))
        require(prior["status"] == "ONE_DEVELOPMENT_LOCAL100_INPUT_BUILT",
                "C125_SUCCESSFUL_C120_REQUIRED")
        candidate = prior["candidate_id"]
        question_hash = prior["question_sha256"]
        expected = prior["hybrid100_order"]
        require(len(expected) == len(set(expected)) == 100,
                "C125_INVALID_REFERENCE_ORDER")
        result.update(candidate_id=candidate, question_sha256=question_hash)

        stage = "CHECK_PREPARED_RUNTIME_INPUTS"
        scope = dict(globals())
        required_names = (
            "np", "collections", "math", "qdrant_client", "DENSE_COLLECTION_NAME",
            "frozen_sparse_index", "analyze_frozen_sparse_text",
            "C125_BOUND_DEVELOPMENT_QUERY",
        )
        missing = [name for name in required_names if scope.get(name) is None]
        result["missing_runtime_bindings"] = missing
        require(not missing, "C125_PREPARED_RUNTIME_INPUTS_REQUIRED")

        # This binding must come from the existing, verified Development cache.
        query = scope["C125_BOUND_DEVELOPMENT_QUERY"]
        question = query["question"]
        require(
            query["candidate_id"] == candidate
            and sha(question.encode("utf-8")) == question_hash,
            "C125_QUERY_IDENTITY_CONFLICT",
        )
        np_local = scope["np"]
        vector = np_local.asarray(query["query_vector"])
        require(
            vector.ndim == 1 and vector.size > 0
            and bool(np_local.isfinite(vector).all()),
            "C125_INVALID_QUERY_VECTOR",
        )
        result["query_vector_sha256"] = sha(vector.tobytes())
        result["runtime_provenance_verified_by_this_cell"] = False

        stage = "LOAD_PINNED_FUNCTIONS"
        names = {
            "frozen_bm25_scores", "frozen_sparse_retrieve", "frozen_dense_retrieve",
            "frozen_minmax_scores", "frozen_weighted_fusion",
        }
        definitions = [
            node for node in tree.body
            if isinstance(node, ast.FunctionDef) and node.name in names
        ]
        require(len(definitions) == len(names), "C125_PINNED_DEFINITIONS_MISSING")
        scope.update(
            FROZEN_CANDIDATE_DEPTH=100,
            FROZEN_DENSE_WEIGHT=0.4,
            FROZEN_SPARSE_WEIGHT=0.6,
        )
        exec(compile(ast.Module(body=definitions, type_ignores=[]), nb_path, "exec"), scope)

        stage = "RUN_ONE_HYBRID_QUERY"
        sparse = scope["frozen_sparse_retrieve"](question, 100)
        result["qdrant_queries_attempted"] += 1
        dense = scope["frozen_dense_retrieve"](vector, 100)
        fused = scope["frozen_weighted_fusion"](
            dense["ranked_chunk_ids"], dense["ranked_scores"],
            sparse["ranked_chunk_ids"], sparse["ranked_scores"],
        )
        observed = fused["ranked_chunk_ids"][:100]
        first_difference = next(
            (
                {"rank": rank + 1, "observed": observed[rank] if rank < len(observed) else None,
                 "expected": cid}
                for rank, cid in enumerate(expected)
                if rank >= len(observed) or observed[rank] != cid
            ),
            None,
        )
        matched = observed == expected
        result.update(
            status="HYBRID100_ORDER_MATCHED" if matched else "HYBRID100_ORDER_DIFFERS",
            full_hybrid100_order_matches=matched,
            observed_depth=len(observed),
            first_difference=first_difference,
            blocker=None,
        )
    except Exception as error:
        result.update(
            status="HYBRID100_PROBE_BLOCKED",
            blocker={"stage": stage, "type": type(error).__name__, "detail": str(error)},
        )

    raw = json.dumps(result, ensure_ascii=False, sort_keys=True, indent=2).encode("utf-8")
    digest = sha(raw)
    relative = reports / f"C125.R11_hybrid100_probe.{digest}.json"
    target = root / relative
    try:
        with target.open("xb") as handle:
            handle.write(raw)
    except FileExistsError:
        pass
    require(target.read_bytes() == raw, "C125_REPORT_RELOAD_CONFLICT")
    result["output"] = {"path": relative.as_posix(), "file_bytes_sha256": digest,
                        "reload_validation": "PASS"}
    print(json.dumps(result, ensure_ascii=False, indent=2))
    return result


V3_C125 = run_v3_c125()
````

### سل ۱۲۶ — آزمون Hybrid100 با دارایی‌های محلی

</div>

````python
# C126 — R11.3 — Load pinned NB16 assets and compare one Hybrid100 order
def run_v3_c126():
    def require(ok, code):
        if not ok:
            raise ValueError(code)

    def sha(raw):
        return hashlib.sha256(raw).hexdigest()

    reports = Path("reports/phase2/quality_improvement/NB22E.lab.v3")
    root = next(
        (p for p in (Path.cwd(), *Path.cwd().parents) if (p / reports).is_dir()),
        None,
    )
    require(root is not None, "C126_PROJECT_ROOT_NOT_FOUND")
    root = root.resolve()

    def read_bound(relative, digest):
        path = (root / relative).resolve()
        require(path.is_relative_to(root), "C126_PATH_OUTSIDE_PROJECT")
        raw = path.read_bytes()
        require(sha(raw) == digest, f"C126_HASH_CONFLICT:{relative}")
        return raw

    nb_path = "notebooks/phase2/16_hybrid_retrieval.ipynb"
    nb_sha = "40699ca54c57fcef40472df6b6fc42a54f74dc2e9bdc9bb40e0ccc93bb7ad6a7"
    c120_sha = "33578338372385982a433939fbec3e140cac3495d0ef8228847cceb99ac05975"
    result = {
        "cell": "C126",
        "roadmap_action": "R11",
        "roadmap_subtask": "R11.3",
        "scope": "ONE_DEVELOPMENT_QUERY_LOCAL_NB16_ASSETS",
        "input_file_hashes": {"NB16": nb_sha, "C120": c120_sha},
        "full_hybrid100_order_matches": False,
        "product_runtime_binding_verified": False,
        "product_regression_tests_executed": 0,
        "new_api_calls": 0,
        "embedding_calls": 0,
        "qdrant_queries_attempted": 0,
        "target_case_rankings_reconstructed": 0,
        "rank_reconstruction_for_unverified_candidates": "DEFERRED",
        "done_item_count": 7,
        "open_item_count": 5,
        "newly_completed_items_from_previous_open_set": [],
        "historical_gate": "FAIL",
        "selected_version_gate": "BLOCKED",
    }

    # NB16 runs in an isolated namespace with dependencies from C001.
    scope = dict(globals())
    scope.pop("qdrant_client", None)
    scope.pop("development_cache_hit", None)
    stage = "VERIFY_PINNED_INPUTS"

    try:
        notebook = json.loads(read_bound(nb_path, nb_sha))
        prior = json.loads(read_bound(
            reports / f"C120.R11_development_local100_input.{c120_sha}.json",
            c120_sha,
        ))
        expected = prior["hybrid100_order"]
        candidate = prior["candidate_id"]
        require(
            prior["status"] == "ONE_DEVELOPMENT_LOCAL100_INPUT_BUILT"
            and len(expected) == len(set(expected)) == 100,
            "C126_INVALID_C120_REFERENCE",
        )
        result["candidate_id"] = candidate
        result["question_sha256"] = prior["question_sha256"]

        def parse_cell(index):
            cell = notebook["cells"][index]
            require(cell["cell_type"] == "code", f"C126_EXPECTED_CODE_CELL:{index}")
            return ast.parse("".join(cell["source"]), filename=f"{nb_path}:{index}")

        def execute_nodes(index, nodes):
            module = ast.Module(body=nodes, type_ignores=[])
            exec(compile(module, f"{nb_path}:{index}", "exec", optimize=0), scope)

        # Run the reviewed, read-only setup cells. Defer opening Qdrant
        # until the existing Development embedding cache has passed.
        for index in (3, 5, 7, 9, 11, 13):
            stage = f"NB16_SETUP_CELL_{index}"
            execute_nodes(index, parse_cell(index).body)

        require(
            Path(scope["PROJECT_ROOT"]).resolve() == root,
            "C126_NB16_PROJECT_ROOT_CONFLICT",
        )

        # The leading assignments in cell 15 define the dense contract.
        # The client creation and collection checks run later.
        stage = "READ_DENSE_RUNTIME_CONTRACT"
        dense_setup = parse_cell(15).body
        client_index = next(
            i for i, node in enumerate(dense_setup)
            if isinstance(node, ast.Assign)
            and any(isinstance(target, ast.Name) and target.id == "qdrant_client"
                    for target in node.targets)
        )
        execute_nodes(15, dense_setup[:client_index])

        stage = "LOAD_EXISTING_DEVELOPMENT_CACHE"
        cache_nodes = parse_cell(17).body
        fallback = [
            i for i, node in enumerate(cache_nodes)
            if isinstance(node, ast.If)
            and isinstance(node.test, ast.UnaryOp)
            and isinstance(node.test.op, ast.Not)
            and isinstance(node.test.operand, ast.Name)
            and node.test.operand.id == "development_cache_hit"
        ]
        require(len(fallback) == 1, "C126_CACHE_FALLBACK_BOUNDARY_CONFLICT")
        # Execute only the source up to the branch that builds a new cache.
        execute_nodes(17, cache_nodes[:fallback[0]])
        require(
            scope["development_cache_hit"] is True,
            "C126_EXISTING_DEVELOPMENT_CACHE_REQUIRED",
        )

        ids = scope["development_candidate_ids_ordered"]
        require(len(ids) == len(set(ids)) == 2664 and candidate in ids,
                "C126_DEVELOPMENT_QUERY_POPULATION_CONFLICT")
        question = scope["development_query_by_candidate_id"][candidate]
        require(
            sha(question.strip().encode("utf-8")) == prior["question_sha256"],
            "C126_QUESTION_HASH_CONFLICT",
        )
        # NB16 originally embedded and analyzed the unstripped Golden text.
        row_index = ids.index(candidate)
        vectors = scope["development_query_embeddings"]
        np_local = scope["np"]
        require(vectors.shape == (len(ids), 384) and vectors.dtype == np_local.float32,
                "C126_CACHE_ARRAY_CONTRACT_CONFLICT")
        vector = vectors[row_index]
        require(bool(np_local.isfinite(vector).all())
                and bool(np_local.isclose(np_local.linalg.norm(vector), 1.0)),
                "C126_INVALID_QUERY_VECTOR")
        result.update(
            cache_row_index_zero_based=row_index,
            cache_key=scope["query_embedding_cache_key"],
            exact_nb16_question_sha256=sha(question.encode("utf-8")),
            query_vector_sha256=sha(vector.tobytes()),
        )
        for key in ("DEVELOPMENT_QUERY_EMBEDDINGS_PATH", "DEVELOPMENT_QUERY_METADATA_PATH"):
            path = Path(scope[key]).resolve()
            result["input_file_hashes"][key] = sha(path.read_bytes())

        stage = "OPEN_AND_VERIFY_LOCAL_QDRANT"
        dense_path = Path(scope["dense_storage_path"]).resolve()
        require(dense_path.is_relative_to(root) and dense_path.is_dir(),
                "C126_LOCAL_QDRANT_PATH_REQUIRED")
        execute_nodes(15, dense_setup[client_index:])
        result.update(
            dense_collection=scope["DENSE_COLLECTION_NAME"],
            dense_storage_path=dense_path.relative_to(root).as_posix(),
        )

        stage = "LOAD_PINNED_HYBRID_FUNCTIONS"
        names = {
            "frozen_bm25_scores", "frozen_sparse_retrieve", "frozen_dense_retrieve",
            "frozen_minmax_scores", "frozen_weighted_fusion",
        }
        definitions = [
            node for node in parse_cell(37).body
            if isinstance(node, ast.FunctionDef) and node.name in names
        ]
        require(len(definitions) == len(names), "C126_PINNED_DEFINITIONS_MISSING")
        scope.update(
            FROZEN_CANDIDATE_DEPTH=100,
            FROZEN_DENSE_WEIGHT=0.4,
            FROZEN_SPARSE_WEIGHT=0.6,
        )
        execute_nodes(37, definitions)

        stage = "RUN_AND_COMPARE_ONE_HYBRID_QUERY"
        sparse = scope["frozen_sparse_retrieve"](question, 100)
        result["qdrant_queries_attempted"] += 1
        dense = scope["frozen_dense_retrieve"](vector, 100)
        fused = scope["frozen_weighted_fusion"](
            dense["ranked_chunk_ids"], dense["ranked_scores"],
            sparse["ranked_chunk_ids"], sparse["ranked_scores"],
        )
        observed = fused["ranked_chunk_ids"][:100]
        differences = [
            {"rank": i + 1, "expected": cid,
             "observed": observed[i] if i < len(observed) else None}
            for i, cid in enumerate(expected)
            if i >= len(observed) or observed[i] != cid
        ]
        matched = observed == expected
        result.update(
            status="LOCAL_HYBRID100_ORDER_MATCHED" if matched
                   else "LOCAL_HYBRID100_ORDER_DIFFERS",
            full_hybrid100_order_matches=matched,
            observed_depth=len(observed),
            observed_hybrid100_order=observed,
            different_position_count=len(differences),
            first_difference=differences[0] if differences else None,
            blocker=None,
        )
    except Exception as error:
        result.update(
            status="LOCAL_HYBRID100_PROBE_BLOCKED",
            blocker={
                "stage": stage,
                "type": type(error).__name__,
                "detail": str(error),
                "missing_name": error.name if isinstance(error, NameError) else None,
            },
        )
    finally:
        client = scope.get("qdrant_client")
        if client is not None:
            try:
                client.close()
            except Exception as error:
                result["qdrant_close_error"] = {
                    "type": type(error).__name__, "detail": str(error),
                }

    raw = json.dumps(
        result, ensure_ascii=False, sort_keys=True, indent=2, allow_nan=False
    ).encode("utf-8")
    digest = sha(raw)
    relative = reports / f"C126.R11_local_hybrid100_probe.{digest}.json"
    path = root / relative
    try:
        with path.open("xb") as handle:
            handle.write(raw)
            handle.flush()
            os.fsync(handle.fileno())
    except FileExistsError:
        pass
    require(path.read_bytes() == raw, "C126_REPORT_RELOAD_CONFLICT")
    result["output"] = {
        "path": relative.as_posix(),
        "file_bytes_sha256": digest,
        "reload_validation": "PASS",
    }
    print(json.dumps(
        {k: v for k, v in result.items() if k != "observed_hybrid100_order"},
        ensure_ascii=False, indent=2,
    ))
    return result


V3_C126 = run_v3_c126()
````

### سل ۱۲۷ — اتصال Hybrid100 محلی به مسیر اختیاری محصول

````python
# C127 — R11.3 — Run local Hybrid100 inside the opt-in rank provider
def run_v3_c127():
    def require(ok, code):
        if not ok:
            raise ValueError(code)

    require(callable(globals().get("run_v3_c126")), "C127_C126_DEFINITION_REQUIRED")
    history = globals().get("In")
    require(isinstance(history, list), "C127_IPYTHON_INPUT_HISTORY_REQUIRED")

    # Reuse the exact C124 definition from this kernel. Do not execute its call.
    header = "def run_v3_c124():"
    source = next(
        (entry for entry in reversed(history)
         if isinstance(entry, str) and header in entry.splitlines()),
        None,
    )
    require(source is not None, "C127_C124_SOURCE_NOT_FOUND_IN_KERNEL_HISTORY")
    tree = ast.parse(source)
    definitions = [
        node for node in tree.body
        if isinstance(node, ast.FunctionDef) and node.name == "run_v3_c124"
    ]
    require(len(definitions) == 1, "C127_C124_DEFINITION_AMBIGUOUS")
    source = ast.get_source_segment(source, definitions[0])
    require(bool(source), "C127_C124_SOURCE_EXTRACTION_FAILED")

    # Require each expected patch point to occur exactly once.
    def replace_once(text, old, new):
        require(text.count(old) == 1, f"C127_C124_SOURCE_CONTRACT_CHANGED:{old}")
        return text.replace(old, new, 1)

    source = replace_once(
        source, "            hybrid = local[\"hybrid100_order\"]",
        """            hybrid = c127_local_hybrid(
                candidate_id, question_hash, local["hybrid100_order"], result, events
            )""",
    )
    source = replace_once(
        source,
        '            "provider_verified_full_orders", "input_receipt_written",',
        '            "local_hybrid100_verified",\n'
        '            "provider_verified_full_orders", "input_receipt_written",',
    )
    source = replace_once(
        source, '"ONE_DEVELOPMENT_QUERY_STORED_INPUT_PROVIDERS_STUB_GENERATOR"',
        '"ONE_DEVELOPMENT_QUERY_LOCAL_HYBRID_STORED_LOCAL100_AND_CHAR_STUB_GENERATOR"',
    )
    source = replace_once(
        source, '"scope": "HASH_BOUND_STORED_DEVELOPMENT_INPUTS",',
        '"scope": "LOCAL_HYBRID100_WITH_HASH_BOUND_STORED_LOCAL100_AND_CHAR",\n'
        '                    "local_hybrid100_receipt_sha256": '
        'result["hybrid100_probe_receipt"]["file_bytes_sha256"],',
    )
    source = replace_once(
        source, '"VERSION_BOUND_STORED_INPUT_PROVIDER_PASSED_OPT_IN_PROBE"',
        '"LOCAL_HYBRID100_CONNECTED_TO_OPT_IN_PROVIDER"',
    )
    source = replace_once(
        source, '"BIND_LIVE_VERSIONED_PROVIDERS_AT_PRODUCT_ENTRYPOINT"',
        '"BIND_REMAINING_LIVE_VERSIONED_PROVIDERS_AT_PRODUCT_ENTRYPOINT"',
    )
    source = replace_once(
        source, '            stored_input_opt_in_integration_verified=True,',
        '            local_hybrid100_opt_in_integration_verified=True,',
    )
    source = replace_once(
        source, '        "stored_input_opt_in_integration_verified": False,',
        '        "local_hybrid100_opt_in_integration_verified": False,\n'
        '        "qdrant_queries_attempted": 0,',
    )
    source = replace_once(
        source, '            status="VERSION_BOUND_OPT_IN_PROBE_BLOCKED",',
        '            status="LOCAL_HYBRID100_OPT_IN_INTEGRATION_BLOCKED",',
    )
    source = source.replace("C124", "C127").replace("run_v3_c124", "run_v3_c127_probe")
    source = replace_once(
        source, '"C127.R11_version_bound_provider_probe"',
        '"C127.R11_local_hybrid100_opt_in_integration"',
    )

    calls = []

    def c127_local_hybrid(candidate, question_hash, expected, result, events):
        require(not calls, "C127_MULTIPLE_HYBRID_INVOCATIONS")
        calls.append(candidate)
        # Suppress NB16 setup prints. C126 still persists its own receipt.
        original = globals()["run_v3_c126"]
        quiet_scope = dict(original.__globals__)
        quiet_scope["print"] = lambda *args, **kwargs: None
        quiet_probe = type(original)(original.__code__, quiet_scope)
        probe = quiet_probe()

        result["hybrid100_probe_receipt"] = probe["output"]
        result["qdrant_queries_attempted"] = probe["qdrant_queries_attempted"]
        result["hybrid100_probe_status"] = probe["status"]
        result["hybrid100_probe_blocker"] = probe.get("blocker")
        observed = probe.get("observed_hybrid100_order")
        require(
            probe["status"] == "LOCAL_HYBRID100_ORDER_MATCHED"
            and probe["full_hybrid100_order_matches"] is True
            and probe["candidate_id"] == candidate
            and probe["question_sha256"] == question_hash
            and probe["qdrant_queries_attempted"] == 1
            and probe["embedding_calls"] == 0
            and probe["new_api_calls"] == 0
            and observed == expected,
            "C127_LOCAL_HYBRID100_PROBE_DID_NOT_MATCH",
        )
        events.append("local_hybrid100_verified")
        return observed

    scope = dict(globals())
    scope["c127_local_hybrid"] = c127_local_hybrid
    exec(compile(source, "<C127:adapted-C124>", "exec"), scope)
    return scope["run_v3_c127_probe"]()


V3_C127 = run_v3_c127()
````

### سل ۱۲۸ — اتصال Hybrid100 و Local100 محلی به مسیر اختیاری

در یک فراخوانی مسیر اختیاری، Hybrid100 با یک جست‌وجوی محلی ساخته می‌شود. پس از تطابق کامل آن با مرجع، Local100 با استفاده از ۵۰ امتیاز ذخیره‌شده و امتیازدهی محلی ۵۰ جفت باقی‌مانده ساخته می‌شود. ترتیب تازهٔ Local100 پیش از ورود به Quota100 کنترل می‌شود. هش رسید هر دو مرحله در رسید ورودی مولد آزمایشی ثبت می‌شود.

دامنه یک سؤال Development است. کانال char از فایل نسخه‌بندی‌شده خوانده می‌شود. تعداد فراخوانی API صفر است.

</div>

````python
# C128 — R11.3 — Compute Local100 inside the verified opt-in probe
def run_v3_c128():
    def require(ok, code):
        if not ok:
            raise ValueError(code)

    def replace_once(source, old, new):
        require(source.count(old) == 1, f"C128_SOURCE_CONTRACT_CHANGED:{old}")
        return source.replace(old, new, 1)

    def definition(name):
        history = globals().get("In")
        require(isinstance(history, list), "C128_IPYTHON_HISTORY_REQUIRED")
        header = f"def {name}():"
        entry = next(
            (item for item in reversed(history)
             if isinstance(item, str) and header in item.splitlines()),
            None,
        )
        require(entry is not None, f"C128_SOURCE_NOT_FOUND:{name}")
        matches = [
            node for node in ast.parse(entry).body
            if isinstance(node, ast.FunctionDef) and node.name == name
        ]
        require(len(matches) == 1, f"C128_SOURCE_AMBIGUOUS:{name}")
        return ast.get_source_segment(entry, matches[0])

    local_probe = globals().get("run_v3_c120")
    require(callable(local_probe), "C128_C120_DEFINITION_REQUIRED")
    require(callable(globals().get("run_v3_c126")), "C128_C126_DEFINITION_REQUIRED")

    def c128_local100(hybrid, local, result, events):
        require(hybrid == local["hybrid100_order"], "C128_HYBRID_INPUT_CONFLICT")
        require("local100_probe_receipt" not in result, "C128_DUPLICATE_LOCAL100_CALL")

        # Reuse C120's pinned scoring implementation in this provider call.
        quiet_globals = dict(local_probe.__globals__)
        quiet_globals["print"] = lambda *args, **kwargs: None
        scored = type(local_probe)(local_probe.__code__, quiet_globals)()
        result.update(
            local100_probe_receipt=scored["output"],
            local100_probe_status=scored["status"],
            local100_probe_blocker=scored.get("blocker"),
            new_local_pairs_scored=scored["new_local_pairs_scored"],
            local_model_inference_calls_attempted=scored["model_inference_calls_attempted"],
        )
        require(
            scored["status"] == "ONE_DEVELOPMENT_LOCAL100_INPUT_BUILT"
            and scored["candidate_id"] == local["candidate_id"]
            and scored["question_sha256"] == local["question_sha256"]
            and scored["hybrid100_order"] == hybrid
            and scored["scored_pool50_order"] == local["scored_pool50_order"]
            and scored["reused_pool50_score_count"] == 50
            and scored["new_local_pairs_scored"] == 50
            and scored["new_api_calls"] == 0
            and scored["local100_order"] == local["local100_order"],
            "C128_LOCAL100_ORDER_OR_PROVENANCE_CONFLICT",
        )
        events.append("local100_verified")
        return scored["local100_order"]

    # Keep the C127 checks and adapt only its C124 template and report labels.
    template = definition("run_v3_c124")
    template = replace_once(
        template,
        '            local_order = local["local100_order"]',
        '            local_order = c128_local100(hybrid, local, result, events)',
    )
    template = replace_once(
        template,
        '                    "local100_receipt_sha256": c120_sha,',
        '                    "local100_receipt_sha256": '
        'result["local100_probe_receipt"]["file_bytes_sha256"],',
    )
    template = replace_once(
        template,
        '            status="VERSION_BOUND_STORED_INPUT_PROVIDER_PASSED_OPT_IN_PROBE",',
        '            local100_opt_in_integration_verified=True,\n'
        '            status="VERSION_BOUND_STORED_INPUT_PROVIDER_PASSED_OPT_IN_PROBE",',
    )
    template = replace_once(
        template,
        '        "stored_input_opt_in_integration_verified": False,',
        '        "local100_opt_in_integration_verified": False,\n'
        '        "stored_input_opt_in_integration_verified": False,',
    )

    adapter = definition("run_v3_c127")
    adapter = replace_once(
        adapter,
        """    source = source.replace("C124", "C127").replace("run_v3_c124", "run_v3_c127_probe")""",
        """    source = replace_once(
        source,
        '            "local_hybrid100_verified",\\n',
        '            "local_hybrid100_verified", "local100_verified",\\n',
    )
    source = source.replace("C124", "C128").replace("run_v3_c124", "run_v3_c127_probe")""",
    )
    adapter = adapter.replace("C127.R11_", "C128.R11_")
    for old, new in (
        (
            "ONE_DEVELOPMENT_QUERY_LOCAL_HYBRID_STORED_LOCAL100_AND_CHAR_STUB_GENERATOR",
            "ONE_DEVELOPMENT_QUERY_LOCAL_HYBRID_AND_LOCAL100_STORED_CHAR_STUB_GENERATOR",
        ),
        (
            "LOCAL_HYBRID100_WITH_HASH_BOUND_STORED_LOCAL100_AND_CHAR",
            "LOCAL_HYBRID100_AND_LOCAL100_WITH_HASH_BOUND_STORED_CHAR",
        ),
        (
            "LOCAL_HYBRID100_CONNECTED_TO_OPT_IN_PROVIDER",
            "LOCAL_HYBRID100_AND_LOCAL100_CONNECTED_TO_OPT_IN_PROVIDER",
        ),
        (
            "BIND_REMAINING_LIVE_VERSIONED_PROVIDERS_AT_PRODUCT_ENTRYPOINT",
            "BIND_LOCAL_CHAR_CHANNEL_AT_PRODUCT_ENTRYPOINT",
        ),
    ):
        adapter = replace_once(adapter, old, new)

    scope = dict(globals())
    scope["In"] = [template]
    scope["c128_local100"] = c128_local100
    exec(compile(adapter, "<C128:adapted-C127>", "exec"), scope)
    return scope["run_v3_c127"]()


V3_C128 = run_v3_c128()
````

````python
# C129 — R11.3 — Compute the pinned NB22D char channel inside the opt-in provider
def run_v3_c129():
    def require(ok, code):
        if not ok:
            raise ValueError(code)

    def sha(raw):
        return hashlib.sha256(raw).hexdigest()

    def replace_once(source, old, new):
        require(source.count(old) == 1, f"C129_SOURCE_CONTRACT_CHANGED:{old}")
        return source.replace(old, new, 1)

    history = globals().get("In")
    require(isinstance(history, list), "C129_IPYTHON_HISTORY_REQUIRED")

    def definition(name):
        header = f"def {name}():"
        entry = next(
            (item for item in reversed(history)
             if isinstance(item, str) and header in item.splitlines()),
            None,
        )
        require(entry is not None, f"C129_SOURCE_NOT_FOUND:{name}")
        matches = [
            node for node in ast.parse(entry).body
            if isinstance(node, ast.FunctionDef) and node.name == name
        ]
        require(len(matches) == 1, f"C129_SOURCE_AMBIGUOUS:{name}")
        return ast.get_source_segment(entry, matches[0])

    reports = Path("reports/phase2/quality_improvement/NB22E.lab.v3")
    root = next(
        (p for p in (Path.cwd(), *Path.cwd().parents) if (p / reports).is_dir()),
        None,
    )
    require(root is not None, "C129_PROJECT_ROOT_NOT_FOUND")
    root = root.resolve()

    def read_bound(relative, digest):
        relative = Path(relative)
        path = (root / relative).resolve()
        require(
            not relative.is_absolute() and path.is_relative_to(root),
            "C129_PATH_OUTSIDE_PROJECT",
        )
        raw = path.read_bytes()
        require(sha(raw) == digest, f"C129_HASH_CONFLICT:{relative}")
        return raw

    c128_sha = "1e155eea20dcf63100fa331f0368a0d5c56ceed6f262a9419c95d6239e348af4"
    previous = json.loads(read_bound(
        reports / f"C128.R11_local_hybrid100_opt_in_integration.{c128_sha}.json",
        c128_sha,
    ))
    require(
        previous["local_hybrid100_opt_in_integration_verified"] is True
        and previous["local100_opt_in_integration_verified"] is True,
        "C129_SUCCESSFUL_C128_REQUIRED",
    )

    def c129_char_channel(question, candidate, saved_channel, result):
        result["input_file_hashes"]["C128"] = c128_sha
        require(candidate == previous["candidate_id"], "C129_CANDIDATE_CONFLICT")
        require(
            sha(question.encode("utf-8")) == previous["question_sha256"],
            "C129_QUESTION_CONFLICT",
        )

        notebook_path = "notebooks/phase2/22d_full_corpus_retrieval_expansion.ipynb"
        notebook_sha = "fff5bd072a4977820882045451b61d66f8621ef9a4918775e4534800f0ae1480"
        notebook = json.loads(read_bound(notebook_path, notebook_sha))
        source = "".join(notebook["cells"][12]["source"])
        names = {"normalized_words", "char_tokens", "build_index", "lexical_search"}
        nodes = [
            node for node in ast.parse(source).body
            if isinstance(node, ast.FunctionDef) and node.name in names
        ]
        require(
            len(nodes) == len(names) and {node.name for node in nodes} == names,
            "C129_PINNED_CHAR_DEFINITIONS_MISSING",
        )

        corpus_path = "data/phase2/chunks/nb11_benchmark_v1/baseline_phase1_char_safe_v1.jsonl"
        corpus_sha = previous["input_file_hashes"]["chunks"]
        chunks = {}
        for line in read_bound(corpus_path, corpus_sha).splitlines():
            if line.strip():
                row = json.loads(line)
                cid = row["chunk_id"]
                require(cid not in chunks, "C129_DUPLICATE_CORPUS_CHUNK")
                chunks[cid] = row
        require(len(chunks) == 3488, "C129_CORPUS_COUNT_CONFLICT")

        config = {
            "char_max_df_fraction": 0.30,
            "bm25": {"k1": 1.2, "b": 0.3},
        }
        scope = dict(globals())
        scope.update(CONFIG=config, require=require)
        module = ast.Module(body=nodes, type_ignores=[])
        exec(compile(module, notebook_path, "exec"), scope)
        index = scope["build_index"](chunks, "char")
        observed = scope["lexical_search"](index, question, "char")
        matched = observed == saved_channel
        result["input_file_hashes"]["NB22D_source"] = notebook_sha
        result["char_channel_comparison"] = {
            "observed_depth": len(observed),
            "saved_depth": len(saved_channel),
            "full_order_matches": matched,
            "first_different_rank": next(
                (i + 1 for i in range(max(len(observed), len(saved_channel)))
                 if i >= len(observed) or i >= len(saved_channel)
                 or observed[i] != saved_channel[i]),
                None,
            ),
        }
        require(matched, "C129_CHAR_CHANNEL_FULL_ORDER_CONFLICT")
        return observed

    template = definition("run_v3_c124")
    template = replace_once(
        template,
        '            for order, weight in ((quota, 0.75), (channel, 0.25)):',
        '            fresh_char = c129_char_channel(question, candidate_id, channel, result)\n'
        '            for order, weight in ((quota, 0.75), (fresh_char, 0.25)):',
    )
    template = replace_once(
        template,
        '                    "char_channel_file_sha256": channel_sha,',
        '                    "char_channel_reference_file_sha256": channel_sha,\n'
        '                    "char_channel_source": "COMPUTED_FROM_PINNED_CORPUS",\n'
        '                    "char_channel_notebook_sha256": '
        'result["input_file_hashes"]["NB22D_source"],\n'
        '                    "char_channel_corpus_sha256": '
        'result["input_file_hashes"]["chunks"],',
    )

    adapter = definition("run_v3_c128").replace("C128", "C129")
    for old, new in (
        (
            "ONE_DEVELOPMENT_QUERY_LOCAL_HYBRID_AND_LOCAL100_STORED_CHAR_STUB_GENERATOR",
            "ONE_DEVELOPMENT_QUERY_LOCAL_HYBRID_LOCAL100_AND_CHAR_STUB_GENERATOR",
        ),
        (
            "LOCAL_HYBRID100_AND_LOCAL100_WITH_HASH_BOUND_STORED_CHAR",
            "LOCAL_HYBRID100_LOCAL100_AND_CHAR_FROM_VERSIONED_INPUTS",
        ),
        (
            "LOCAL_HYBRID100_AND_LOCAL100_CONNECTED_TO_OPT_IN_PROVIDER",
            "LOCAL_HYBRID100_LOCAL100_AND_CHAR_CONNECTED_TO_OPT_IN_PROVIDER",
        ),
        (
            "BIND_LOCAL_CHAR_CHANNEL_AT_PRODUCT_ENTRYPOINT",
            "VERIFY_PRODUCT_RUNTIME_BINDING_AND_VERSIONED_POOL50_PROVIDER",
        ),
    ):
        adapter = replace_once(adapter, old, new)

    scope = dict(globals())
    scope["In"] = [template, definition("run_v3_c127")]
    scope["c129_char_channel"] = c129_char_channel
    exec(compile(adapter, "<C129:adapted-C128>", "exec"), scope)
    return scope["run_v3_c128"]()


V3_C129 = run_v3_c129()
````

### سل ۱۳۰ — منشأ Pool50 و اتصال مسیر اختیاری محصول

برای همان سؤال Development، رسید سل ۱۲۹، رسید ورودی آن، گزارش ۰۵۶ و ردیف امتیازهای Pool50 با هش کنترل می‌شوند. عضویت Pool50 با ۵۰ رتبهٔ اول Hybrid100 و ترتیب Pool50 با ورودی Local100 مقایسه می‌شود.

سپس سورس پین‌شدهٔ محصول بررسی می‌شود تا مشخص شود متد پیش‌فرض ask در همین کلاس به ask_with_char_rrf ارجاع دارد یا خیر. نتیجهٔ این بررسی فقط دربارهٔ همین فایل و همین مسیر است. اجرای محصول و Gate نسخهٔ منتخب همچنان بازند.

</div>

````python
# C130 — R11.3 — Verify Pool50 provenance and inspect the default product route
def run_v3_c130():
    def require(ok, code):
        if not ok:
            raise ValueError(code)

    def sha(raw):
        return hashlib.sha256(raw).hexdigest()

    reports = Path("reports/phase2/quality_improvement/NB22E.lab.v3")
    cwd = Path.cwd()
    root = next((p for p in (cwd, *cwd.parents) if (p / reports).is_dir()), None)
    require(root is not None, "C130_PROJECT_ROOT_NOT_FOUND")
    root = root.resolve()

    def read_bound(relative, digest):
        relative = Path(relative)
        path = (root / relative).resolve()
        require(not relative.is_absolute() and path.is_relative_to(root),
                "C130_PATH_OUTSIDE_PROJECT")
        raw = path.read_bytes()
        require(sha(raw) == digest, f"C130_HASH_CONFLICT:{relative}")
        return raw

    c129_sha = "f28bfb9302b4f92ad2408d60f94aef549b515a436fbb54be773b4b9c0a331811"
    result = {
        "cell": "C130",
        "roadmap_action": "R11",
        "roadmap_subtask": "R11.3",
        "scope": "ONE_DEVELOPMENT_QUERY_AND_PINNED_PIPELINE_SOURCE",
        "input_file_hashes": {"C129": c129_sha},
        "pool50_provenance_verified": False,
        "product_runtime_binding_verified": False,
        "done_item_count": 7,
        "open_item_count": 5,
        "open_item_ids": [8, 9, 10, 11, 12],
        "newly_completed_items_from_previous_open_set": [],
        "historical_gate": "FAIL",
        "selected_version_gate": "BLOCKED",
        "new_api_calls": 0,
        "qdrant_queries_attempted": 0,
        "new_local_pairs_scored": 0,
        "target_case_rankings_reconstructed": 0,
        "product_regression_tests_executed": 0,
    }

    stage = "BIND_C129_RECEIPTS"
    try:
        prior = json.loads(read_bound(
            reports / f"C129.R11_local_hybrid100_opt_in_integration.{c129_sha}.json",
            c129_sha,
        ))
        require(
            prior["status"] == "LOCAL_HYBRID100_LOCAL100_AND_CHAR_CONNECTED_TO_OPT_IN_PROVIDER"
            and prior["local_hybrid100_opt_in_integration_verified"] is True
            and prior["local100_opt_in_integration_verified"] is True
            and prior["char_channel_comparison"]["full_order_matches"] is True,
            "C130_SUCCESSFUL_C129_REQUIRED",
        )
        candidate = prior["candidate_id"]
        question_sha = prior["question_sha256"]
        result.update(candidate_id=candidate, question_sha256=question_sha)

        def load_receipt(descriptor, label):
            digest = descriptor["file_bytes_sha256"]
            payload = json.loads(read_bound(descriptor["path"], digest))
            result["input_file_hashes"][label] = digest
            return payload

        receipt = load_receipt(prior["receipts"]["input"], "C129_input")
        local = load_receipt(prior["local100_probe_receipt"], "C120")
        hybrid = load_receipt(prior["hybrid100_probe_receipt"], "C126")
        identity = receipt["runtime_identity"]
        require(
            local["candidate_id"] == hybrid["candidate_id"] == receipt["candidate_id"] == candidate
            and local["question_sha256"] == hybrid["question_sha256"]
                == receipt["question_sha256"] == question_sha
            and receipt["selected_variant"] == "char_rrf"
            and identity["local100_receipt_sha256"] == result["input_file_hashes"]["C120"]
            and identity["local_hybrid100_receipt_sha256"] == result["input_file_hashes"]["C126"]
            and identity["pipeline_source_sha256"] == prior["input_file_hashes"]["pipeline_source"],
            "C130_INPUT_RECEIPT_BINDING_CONFLICT",
        )
        hybrid_order = hybrid["observed_hybrid100_order"]
        require(
            hybrid["full_hybrid100_order_matches"] is True
            and len(hybrid_order) == len(set(hybrid_order)) == 100
            and hybrid_order == local["hybrid100_order"],
            "C130_HYBRID100_ORDER_CONFLICT",
        )

        stage = "VERIFY_SAVED_POOL50_AND_SCORE_BINDING"
        c056_sha = local["input_file_hashes"]["NB22_C056"]
        c056_path = (
            Path("reports/phase2/release/NB22.C056.retrieval_verification")
            / f"C056.independent_retrieval_metric_verification.{c056_sha}.json"
        )
        c056 = json.loads(read_bound(c056_path, c056_sha))
        binding = c056["input_file_receipts"]["reranker"]
        digest = binding["file_bytes_sha256"]
        require(digest == local["input_file_hashes"]["reranker"],
                "C130_RERANKER_SOURCE_PIN_CONFLICT")

        matches = []
        for number, line in enumerate(read_bound(binding["path"], digest).splitlines(), 1):
            if line.strip():
                row = json.loads(line)
                if row["candidate_id"] == candidate:
                    matches.append((number, row))
        require(len(matches) == 1, "C130_EXPECTED_ONE_POOL50_ROW")
        line_number, row = matches[0]
        records = row["candidates"]
        pool = [record["chunk_id"] for record in records]
        require(
            [record["rank"] for record in records] == list(range(1, 51))
            and len(set(pool)) == 50,
            "C130_INVALID_POOL50_RANKS",
        )
        require(set(pool) == set(hybrid_order[:50]), "C130_POOL50_MEMBERSHIP_CONFLICT")
        require(pool == local["scored_pool50_order"], "C130_SAVED_POOL50_ORDER_CONFLICT")
        require(
            local["reused_pool50_score_count"] == 50
            and all(float(record["reranker_score"]) == local["scores_by_chunk_id"][record["chunk_id"]]
                    for record in records),
            "C130_REUSED_POOL50_SCORE_CONFLICT",
        )
        result["input_file_hashes"].update(NB22_C056=c056_sha, saved_pool50=digest)
        result.update(
            pool50_provenance_verified=True,
            pool50_source={
                "path": binding["path"],
                "file_bytes_sha256": digest,
                "line_number": line_number,
            },
            pool50_membership_matches_hybrid_first50=True,
            pool50_order_matches_local100_input=True,
            pool50_order_equals_hybrid_first50=(pool == hybrid_order[:50]),
            reused_pool50_scores_verified=50,
        )

        stage = "INSPECT_PINNED_PRODUCT_ROUTE"
        pipeline_path = "src/pipeline/rag_pipeline.py"
        pipeline_sha = prior["input_file_hashes"]["pipeline_source"]
        tree = ast.parse(read_bound(pipeline_path, pipeline_sha), filename=pipeline_path)
        classes = [node for node in tree.body if isinstance(node, ast.ClassDef)
                   and node.name == "RAGPipeline"]
        require(len(classes) == 1, "C130_RAGPIPELINE_CLASS_CONFLICT")
        methods = {
            node.name: node for node in classes[0].body
            if isinstance(node, ast.FunctionDef)
        }
        require({"ask", "ask_with_char_rrf"} <= methods.keys(),
                "C130_PIPELINE_METHODS_REQUIRED")
        references = [
            node.lineno for node in ast.walk(methods["ask"])
            if (isinstance(node, ast.Attribute) and node.attr == "ask_with_char_rrf")
            or (isinstance(node, ast.Name) and node.id == "ask_with_char_rrf")
            or (isinstance(node, ast.Constant) and node.value == "ask_with_char_rrf")
        ]
        result["input_file_hashes"]["pipeline_source"] = pipeline_sha
        result["default_route_source_check"] = {
            "path": pipeline_path,
            "ask_line": methods["ask"].lineno,
            "ask_with_char_rrf_line": methods["ask_with_char_rrf"].lineno,
            "opt_in_reference_lines_in_ask": sorted(set(references)),
        }
        result.update(
            status="SAVED_POOL50_BOUND_AND_PRODUCT_ROUTE_INSPECTED",
            blocker=None,
            next_product_dependency=(
                "VERIFY_EFFECTIVE_CALLER_AND_POOL50_PROVIDER_CONFIGURATION"
                if references else
                "IDENTIFY_OR_WIRE_PRODUCT_CALLER_FOR_OPT_IN_ROUTE_AND_POOL50_PROVIDER"
            ),
        )
    except (OSError, ValueError, KeyError, TypeError, SyntaxError) as error:
        result.update(
            status="POOL50_OR_ROUTE_CHECK_BLOCKED",
            blocker={"stage": stage, "type": type(error).__name__, "detail": str(error)},
        )

    raw = json.dumps(result, ensure_ascii=False, sort_keys=True, indent=2,
                     allow_nan=False).encode("utf-8")
    digest = sha(raw)
    relative = reports / f"C130.R11_pool50_provenance_and_product_route.{digest}.json"
    target = root / relative
    try:
        with target.open("xb") as handle:
            handle.write(raw)
            handle.flush()
            os.fsync(handle.fileno())
    except FileExistsError:
        pass
    require(target.read_bytes() == raw, "C130_REPORT_RELOAD_CONFLICT")
    result["output"] = {
        "path": relative.as_posix(),
        "file_bytes_sha256": digest,
        "reload_validation": "PASS",
    }
    print(json.dumps(result, ensure_ascii=False, indent=2))
    return result


V3_C130 = run_v3_c130()
````

### سل ۱۳۱ — شناسایی فراخوان مسیر اختیاری محصول

پس از احراز منشأ Pool50 در سل ۱۳۰، فایل‌های Python زیر src، app، apps و scripts و فایل‌های Python ریشه بررسی می‌شوند. تعریف متد از فراخوان مستقیم تفکیک می‌شود. اگر فراخوان مستقیمی در این دامنه پیدا نشود، گام بعد تعیین محل اتصال مسیر اختیاری است.

این بررسی ایستا است. هیچ رتبه‌ای بازسازی نمی‌شود و هیچ فراخوانی API انجام نمی‌شود.

</div>

````python
# C131 — R11.3 — Locate direct callers of the opt-in product route
def run_v3_c131():
    def require(ok, code):
        if not ok:
            raise ValueError(code)

    def sha(raw):
        return hashlib.sha256(raw).hexdigest()

    reports = Path("reports/phase2/quality_improvement/NB22E.lab.v3")
    cwd = Path.cwd()
    root = next((p for p in (cwd, *cwd.parents) if (p / reports).is_dir()), None)
    require(root is not None, "C131_PROJECT_ROOT_NOT_FOUND")
    root = root.resolve()

    c130_sha = "8af22ac2ca7a8b8685c9cbd58bc1fd99dcea781fdd18560417d59edbc03766e0"
    c130_path = reports / f"C130.R11_pool50_provenance_and_product_route.{c130_sha}.json"
    raw = (root / c130_path).read_bytes()
    require(sha(raw) == c130_sha, "C131_C130_HASH_CONFLICT")
    prior = json.loads(raw)
    require(
        prior["status"] == "SAVED_POOL50_BOUND_AND_PRODUCT_ROUTE_INSPECTED"
        and prior["pool50_provenance_verified"] is True
        and prior["reused_pool50_scores_verified"] == 50,
        "C131_SUCCESSFUL_C130_REQUIRED",
    )

    pipeline_rel = prior["default_route_source_check"]["path"]
    pipeline_sha = prior["input_file_hashes"]["pipeline_source"]
    require(
        sha((root / pipeline_rel).read_bytes()) == pipeline_sha,
        "C131_PIPELINE_SOURCE_CHANGED_SINCE_C130",
    )

    method = "ask_with_char_rrf"
    folders = [root / name for name in ("src", "app", "apps", "scripts")]
    active_folders = [folder for folder in folders if folder.is_dir()]
    paths = set(root.glob("*.py"))
    for folder in active_folders:
        paths.update(folder.rglob("*.py"))

    definitions = []
    direct_calls = []
    other_references = []
    skipped = []
    scanned = 0

    def named_reference(node):
        return (
            isinstance(node, ast.Attribute) and node.attr == method
        ) or (
            isinstance(node, ast.Name) and node.id == method
        )

    for path in sorted(paths):
        relative = path.relative_to(root).as_posix()
        if not path.resolve().is_relative_to(root):
            skipped.append({"path": relative, "reason": "PATH_OUTSIDE_PROJECT"})
            continue
        try:
            source = path.read_bytes()
            scanned += 1
            if method.encode("ascii") not in source:
                continue
            tree = ast.parse(source, filename=relative)
        except (OSError, SyntaxError, UnicodeError) as error:
            skipped.append({
                "path": relative, "reason": type(error).__name__,
                "line": getattr(error, "lineno", None),
            })
            continue

        digest = sha(source)
        nodes = list(ast.walk(tree))
        call_targets = set()
        for node in nodes:
            if isinstance(node, (ast.FunctionDef, ast.AsyncFunctionDef)) and node.name == method:
                definitions.append({
                    "path": relative, "line": node.lineno, "file_bytes_sha256": digest,
                })
            if isinstance(node, ast.Call) and named_reference(node.func):
                call_targets.add(id(node.func))
                direct_calls.append({
                    "path": relative,
                    "line": node.lineno,
                    "file_bytes_sha256": digest,
                    "callee": ast.unparse(node.func),
                    "explicit_keyword_names": [
                        keyword.arg for keyword in node.keywords if keyword.arg is not None
                    ],
                    "has_keyword_unpacking": any(
                        keyword.arg is None for keyword in node.keywords
                    ),
                })
        for node in nodes:
            if id(node) in call_targets:
                continue
            if named_reference(node) or (
                isinstance(node, ast.Constant) and node.value == method
            ):
                other_references.append({
                    "path": relative, "line": node.lineno, "file_bytes_sha256": digest,
                })

    if direct_calls:
        status = "DIRECT_OPT_IN_CALL_SITES_FOUND"
        next_step = "VERIFY_CALLER_AND_RANK_SOURCES_BINDING"
    elif skipped or other_references:
        status = "NO_DIRECT_CALL_FOUND_INDIRECT_OR_UNREAD_LEADS_REMAIN"
        next_step = "RESOLVE_RECORDED_LEADS_BEFORE_CHOOSING_INTEGRATION_POINT"
    else:
        status = "NO_DIRECT_OPT_IN_CALL_IN_SCANNED_PYTHON_SOURCES"
        next_step = "LOCATE_PRODUCT_ENTRYPOINT_TO_WIRE_OPT_IN"

    result = {
        "cell": "C131",
        "roadmap_action": "R11",
        "roadmap_subtask": "R11.3",
        "status": status,
        "input_file_hashes": {"C130": c130_sha, "pipeline_source": pipeline_sha},
        "scope": "STATIC_PYTHON_CALL_SITES_IN_DECLARED_PATHS",
        "scanned_directories": [p.name for p in active_folders],
        "root_python_files_included": True,
        "scanned_python_file_count": scanned,
        "definitions": definitions,
        "direct_calls": direct_calls,
        "other_references": other_references,
        "skipped_files": skipped,
        "saved_pool50_source": prior["pool50_source"],
        "product_runtime_binding_verified": False,
        "product_regression_tests_executed": 0,
        "target_case_rankings_reconstructed": 0,
        "new_api_calls": 0,
        "done_item_count": 7,
        "open_item_count": 5,
        "newly_completed_items_from_previous_open_set": [],
        "historical_gate": "FAIL",
        "selected_version_gate": "BLOCKED",
        "next_product_dependency": next_step,
    }

    payload = json.dumps(
        result, ensure_ascii=False, sort_keys=True, indent=2, allow_nan=False
    ).encode("utf-8")
    digest = sha(payload)
    relative = reports / f"C131.R11_opt_in_caller_trace.{digest}.json"
    target = root / relative
    try:
        with target.open("xb") as handle:
            handle.write(payload)
            handle.flush()
            os.fsync(handle.fileno())
    except FileExistsError:
        pass
    require(target.read_bytes() == payload, "C131_REPORT_RELOAD_CONFLICT")
    result["output"] = {
        "path": relative.as_posix(),
        "file_bytes_sha256": digest,
        "reload_validation": "PASS",
    }
    print(json.dumps(result, ensure_ascii=False, indent=2))
    return result


V3_C131 = run_v3_c131()
````

### سل ۱۳۲ — پیدا کردن محل فراخوان مسیر محصول در نوت‌بوک‌ها

رسید سل ۱۳۱ با هش کنترل می‌شود. سپس فقط سل‌های کد نوت‌بوک‌های نسخه‌بندی‌شده برای ساخت RAGPipeline و فراخوان ask یا ask_with_char_rrf بررسی می‌شوند. مسیر نوت‌بوک، شمارهٔ سل، خط و نام فراخوان ثبت می‌شود.

````python
# C132 — R11.3 — Locate notebook callers without executing notebook code
def run_v3_c132():
    def require(ok, code):
        if not ok:
            raise ValueError(code)

    def sha(raw):
        return hashlib.sha256(raw).hexdigest()

    reports = Path("reports/phase2/quality_improvement/NB22E.lab.v3")
    cwd = Path.cwd()
    root = next((p for p in (cwd, *cwd.parents) if (p / reports).is_dir()), None)
    require(root is not None, "C132_PROJECT_ROOT_NOT_FOUND")
    root = root.resolve()

    c131_sha = "dc90f50faf881fa4d54b07743651067c1d2bf2c089ff5825e2ad328762f9888f"
    previous = root / reports / f"C131.R11_opt_in_caller_trace.{c131_sha}.json"
    raw = previous.read_bytes()
    require(sha(raw) == c131_sha, "C132_C131_HASH_CONFLICT")
    prior = json.loads(raw)
    require(
        prior["status"] == "NO_DIRECT_OPT_IN_CALL_IN_SCANNED_PYTHON_SOURCES"
        and not prior["skipped_files"],
        "C132_C131_CALLER_TRACE_REQUIRED",
    )

    listing = subprocess.run(
        ["git", "ls-files", "-z", "--", "notebooks"],
        cwd=root, check=True, capture_output=True,
    )
    paths = [
        Path(os.fsdecode(name)) for name in listing.stdout.split(b"\0")
        if name.endswith(b".ipynb")
    ]
    calls = []
    unread = []
    relevant_unparsed_cells = []
    scanned = 0
    for relative in paths:
        try:
            path = (root / relative).resolve()
            require(path.is_relative_to(root), "C132_PATH_OUTSIDE_PROJECT")
            raw = path.read_bytes()
            cells = json.loads(raw)["cells"]
        except (OSError, ValueError, KeyError, TypeError) as error:
            unread.append({"path": relative.as_posix(), "reason": type(error).__name__})
            continue

        scanned += 1
        digest = sha(raw)
        for index, cell in enumerate(cells):
            if cell.get("cell_type") != "code":
                continue
            source = "".join(cell.get("source", []))
            if not any(token in source for token in (
                "RAGPipeline", "ask_with_char_rrf", ".ask",
            )):
                continue
            try:
                tree = ast.parse(source)
            except SyntaxError as error:
                relevant_unparsed_cells.append({
                    "path": relative.as_posix(),
                    "cell_index_zero_based": index,
                    "line": error.lineno,
                    "file_bytes_sha256": digest,
                })
                continue

            for node in ast.walk(tree):
                if not isinstance(node, ast.Call):
                    continue
                callee = node.func
                name = (
                    callee.attr if isinstance(callee, ast.Attribute)
                    else callee.id if isinstance(callee, ast.Name)
                    else None
                )
                if name not in {"RAGPipeline", "ask", "ask_with_char_rrf"}:
                    continue
                calls.append({
                    "path": relative.as_posix(),
                    "cell_index_zero_based": index,
                    "line": node.lineno,
                    "callee": ast.unparse(callee),
                    "name": name,
                    "keyword_names": [kw.arg for kw in node.keywords if kw.arg],
                    "file_bytes_sha256": digest,
                })

    opt_in = [entry for entry in calls if entry["name"] == "ask_with_char_rrf"]
    constructors = [entry for entry in calls if entry["name"] == "RAGPipeline"]
    asks = [entry for entry in calls if entry["name"] == "ask"]
    result = {
        "cell": "C132",
        "roadmap_action": "R11",
        "roadmap_subtask": "R11.3",
        "status": "TRACKED_NOTEBOOK_CALLER_LEADS_RECORDED",
        "input_file_hashes": {"C131": c131_sha},
        "scope": "STATIC_TRACKED_NOTEBOOK_CODE_CELLS",
        "scanned_notebook_count": scanned,
        "opt_in_call_leads": opt_in,
        "pipeline_constructor_leads": constructors,
        "ask_call_leads": asks,
        "unread_notebooks": unread,
        "relevant_unparsed_cells": relevant_unparsed_cells,
        "next_product_dependency": (
            "VERIFY_OPT_IN_CALLER_AND_PROVIDER_BINDINGS" if opt_in else
            "INSPECT_PIPELINE_CONSTRUCTOR_AND_ASK_LEADS_FOR_INTEGRATION_POINT"
        ),
        "product_runtime_binding_verified": False,
        "product_regression_tests_executed": 0,
        "target_case_rankings_reconstructed": 0,
        "new_api_calls": 0,
        "done_item_count": 7,
        "open_item_count": 5,
        "newly_completed_items_from_previous_open_set": [],
        "historical_gate": "FAIL",
        "selected_version_gate": "BLOCKED",
    }

    payload = json.dumps(
        result, ensure_ascii=False, sort_keys=True, indent=2, allow_nan=False
    ).encode("utf-8")
    digest = sha(payload)
    relative = reports / f"C132.R11_notebook_caller_trace.{digest}.json"
    target = root / relative
    try:
        with target.open("xb") as handle:
            handle.write(payload)
            handle.flush()
            os.fsync(handle.fileno())
    except FileExistsError:
        pass
    require(target.read_bytes() == payload, "C132_REPORT_RELOAD_CONFLICT")

    result["output"] = {
        "path": relative.as_posix(),
        "file_bytes_sha256": digest,
        "reload_validation": "PASS",
    }
    print(json.dumps(result, ensure_ascii=False, indent=2))
    return result


V3_C132 = run_v3_c132()
````

### سل ۱۳۴ — فراخوان صریح مسیر اختیاری محصول

تابع اتصال، نمونهٔ RAGPipeline و فراهم‌کننده‌های Hybrid100، Pool50 و رتبه‌بندی نهایی را دریافت می‌کند. هویت سؤال و عضویت Pool50 کنترل می‌شوند. ورودی‌های احرازشده به رتبه‌بند داده می‌شوند و متد محصول ثبت رسید پیش از تولید و پس از تولید را انجام می‌دهد.

اجرای این سل فقط تابع را تعریف می‌کند. برای آزمون واقعی، فراهم‌کننده‌های نسخه‌بندی‌شده باید در محل فراخوان به آن متصل شوند.

</div>

````python
# C134 — R11.3 — Explicit caller for the product opt-in route
def ask_product_char_rrf(
    pipeline, *, question, candidate_id, hybrid100_provider,
    pool50_provider, rank_provider, write_receipt, top_k=10,
):
    def require(ok, code):
        if not ok:
            raise ValueError(code)

    require(isinstance(question, str) and bool(question.strip()),
            "C134_QUESTION_REQUIRED")
    require(isinstance(candidate_id, str) and bool(candidate_id.strip()),
            "C134_CANDIDATE_ID_REQUIRED")

    question_sha = hashlib.sha256(question.encode("utf-8")).hexdigest()

    def check_identity(payload, label):
        require(
            payload["candidate_id"] == candidate_id
            and payload["question_sha256"] == question_sha,
            f"C134_QUERY_IDENTITY_CONFLICT:{label}",
        )
        digest = payload["receipt_sha256"]
        require(
            isinstance(digest, str) and len(digest) == 64
            and all(char in "0123456789abcdef" for char in digest),
            f"C134_PROVIDER_RECEIPT_HASH_REQUIRED:{label}",
        )
        return digest

    def rank_sources(*, question, candidate_id, top_k):
        hybrid = hybrid100_provider(question=question, candidate_id=candidate_id)
        hybrid_sha = check_identity(hybrid, "HYBRID100")
        hybrid_ids = hybrid["ordered_chunk_ids"]
        require(len(hybrid_ids) == len(set(hybrid_ids)) == 100,
                "C134_INVALID_HYBRID100")

        pool = pool50_provider(
            question=question, candidate_id=candidate_id,
            hybrid100=tuple(hybrid_ids),
        )
        pool_sha = check_identity(pool, "POOL50")
        pool_ids = pool["ordered_chunk_ids"]
        scores = pool["scores_by_chunk_id"]
        require(
            len(pool_ids) == len(set(pool_ids)) == 50
            and set(pool_ids) == set(hybrid_ids[:50]) == set(scores)
            and all(type(score) in (int, float) and math.isfinite(score)
                    for score in scores.values()),
            "C134_POOL50_MEMBERSHIP_OR_SCORES_CONFLICT",
        )

        ranked = rank_provider(
            question=question, candidate_id=candidate_id, top_k=top_k,
            hybrid100=tuple(hybrid_ids),
            pool50_order=tuple(pool_ids),
            pool50_scores=dict(scores),
        )
        return {
            **ranked,
            "runtime_identity": {
                **ranked["runtime_identity"],
                "hybrid100_provider_receipt_sha256": hybrid_sha,
                "pool50_provider_receipt_sha256": pool_sha,
            },
        }

    return pipeline.ask_with_char_rrf(
        question, candidate_id=candidate_id, rank_sources=rank_sources,
        write_receipt=write_receipt, top_k=top_k,
    )


print("C134: OPT_IN_CALLER_DEFINED; PRODUCT_EXECUTION_PENDING")
````

### سل ۱۳۵ — فراهم‌کنندهٔ Pool50 برای فراخوان اختیاری

سل فراهم‌کننده را روی ورودی پین‌شده آزمایش می‌کند. هیچ جست‌وجو، امتیازدهی تازه یا فراخوانی API انجام نمی‌شود.

````python
# C135 — R11.3 — Bind and exercise the versioned Pool50 provider
def run_v3_c135():
    def require(ok, code):
        if not ok:
            raise ValueError(code)

    def sha(raw):
        return hashlib.sha256(raw).hexdigest()

    reports = Path("reports/phase2/quality_improvement/NB22E.lab.v3")
    cwd = Path.cwd()
    root = next((p for p in (cwd, *cwd.parents) if (p / reports).is_dir()), None)
    require(root is not None, "C135_PROJECT_ROOT_NOT_FOUND")
    root = root.resolve()

    def read_bound(relative, digest):
        relative = Path(relative)
        path = (root / relative).resolve()
        require(not relative.is_absolute() and path.is_relative_to(root),
                "C135_PATH_OUTSIDE_PROJECT")
        raw = path.read_bytes()
        require(sha(raw) == digest, f"C135_HASH_CONFLICT:{relative}")
        return raw

    def save(stem, value):
        raw = json.dumps(
            value, ensure_ascii=False, sort_keys=True, indent=2, allow_nan=False,
        ).encode("utf-8")
        digest = sha(raw)
        relative = reports / f"{stem}.{digest}.json"
        path = root / relative
        try:
            with path.open("xb") as handle:
                handle.write(raw)
                handle.flush()
                os.fsync(handle.fileno())
        except FileExistsError:
            pass
        require(path.read_bytes() == raw, "C135_RECEIPT_RELOAD_CONFLICT")
        return {"path": relative.as_posix(), "file_bytes_sha256": digest,
                "reload_validation": "PASS"}

    c130_sha = "8af22ac2ca7a8b8685c9cbd58bc1fd99dcea781fdd18560417d59edbc03766e0"
    prior = json.loads(read_bound(
        reports / f"C130.R11_pool50_provenance_and_product_route.{c130_sha}.json",
        c130_sha,
    ))
    require(prior["pool50_provenance_verified"] is True
            and prior["reused_pool50_scores_verified"] == 50,
            "C135_SUCCESSFUL_C130_REQUIRED")

    candidate = prior["candidate_id"]
    question_sha = prior["question_sha256"]
    pins = prior["input_file_hashes"]
    local = json.loads(read_bound(
        reports / f"C120.R11_development_local100_input.{pins['C120']}.json",
        pins["C120"],
    ))
    require(
        local["candidate_id"] == candidate and local["question_sha256"] == question_sha,
        "C135_C120_QUERY_BINDING_CONFLICT",
    )
    expected_hybrid = tuple(local["hybrid100_order"])
    require(len(expected_hybrid) == len(set(expected_hybrid)) == 100,
            "C135_INVALID_HYBRID100")

    source = prior["pool50_source"]
    require(source["file_bytes_sha256"] == pins["saved_pool50"],
            "C135_POOL50_SOURCE_PIN_CONFLICT")
    lines = read_bound(source["path"], source["file_bytes_sha256"]).splitlines()
    line_number = source["line_number"]
    require(type(line_number) is int and 1 <= line_number <= len(lines),
            "C135_INVALID_POOL50_LINE")
    row_bytes = lines[line_number - 1]
    row = json.loads(row_bytes)
    require(row["candidate_id"] == candidate, "C135_POOL50_CANDIDATE_CONFLICT")

    records = row["candidates"]
    pool_ids = tuple(record["chunk_id"] for record in records)
    scores = {
        record["chunk_id"]: float(record["reranker_score"]) for record in records
    }
    require(
        [record["rank"] for record in records] == list(range(1, 51))
        and len(set(pool_ids)) == 50
        and set(pool_ids) == set(expected_hybrid[:50])
        and list(pool_ids) == local["scored_pool50_order"]
        and all(math.isfinite(score) and score == local["scores_by_chunk_id"][cid]
                for cid, score in scores.items()),
        "C135_POOL50_RANK_OR_SCORE_CONFLICT",
    )

    provider_receipt = save("C135.R11_pool50_provider_receipt", {
        "scope": "ONE_DEVELOPMENT_QUERY_SAVED_RERANKER_SCORES",
        "candidate_id": candidate,
        "question_sha256": question_sha,
        "c130_file_bytes_sha256": c130_sha,
        "pool50_source": source,
        "source_row_sha256": sha(row_bytes),
        "hybrid100_order": list(expected_hybrid),
        "ordered_chunk_ids": list(pool_ids),
        "scores_by_chunk_id": scores,
    })

    def pool50_provider(*, question, candidate_id, hybrid100):
        require(
            candidate_id == candidate
            and isinstance(question, str)
            and sha(question.encode("utf-8")) == question_sha,
            "C135_PROVIDER_QUERY_CONFLICT",
        )
        require(tuple(hybrid100) == expected_hybrid,
                "C135_PROVIDER_HYBRID100_ORDER_CONFLICT")
        return {
            "candidate_id": candidate,
            "question_sha256": question_sha,
            "ordered_chunk_ids": list(pool_ids),
            "scores_by_chunk_id": dict(scores),
            "receipt_sha256": provider_receipt["file_bytes_sha256"],
        }

    c056_sha = pins["NB22_C056"]
    c056 = json.loads(read_bound(
        Path("reports/phase2/release/NB22.C056.retrieval_verification")
        / f"C056.independent_retrieval_metric_verification.{c056_sha}.json",
        c056_sha,
    ))
    golden = c056["input_file_receipts"]["golden"]
    require(golden["file_bytes_sha256"] == local["input_file_hashes"]["golden"],
            "C135_GOLDEN_PIN_CONFLICT")
    questions = []
    for line in read_bound(golden["path"], golden["file_bytes_sha256"]).splitlines():
        if line.strip():
            record = json.loads(line)
            if record["candidate_id"] == candidate:
                questions.append(record["evaluation_content"]["question"].strip())
    require(len(questions) == 1, "C135_EXPECTED_ONE_QUESTION")

    supplied = pool50_provider(
        question=questions[0], candidate_id=candidate, hybrid100=expected_hybrid,
    )
    require(
        supplied["ordered_chunk_ids"] == list(pool_ids)
        and supplied["scores_by_chunk_id"] == scores
        and supplied["receipt_sha256"] == provider_receipt["file_bytes_sha256"],
        "C135_PROVIDER_OUTPUT_CONFLICT",
    )

    # Membership stays identical in this control, but the ordering changes.
    wrong_order = list(expected_hybrid)
    wrong_order[0], wrong_order[1] = wrong_order[1], wrong_order[0]
    try:
        pool50_provider(
            question=questions[0], candidate_id=candidate, hybrid100=wrong_order,
        )
    except ValueError as error:
        require(str(error) == "C135_PROVIDER_HYBRID100_ORDER_CONFLICT",
                "C135_UNEXPECTED_NEGATIVE_CONTROL_ERROR")
    else:
        raise ValueError("C135_PROVIDER_ACCEPTED_WRONG_HYBRID100_ORDER")

    result = {
        "cell": "C135",
        "roadmap_action": "R11",
        "roadmap_subtask": "R11.3",
        "status": "VERSIONED_POOL50_PROVIDER_READY_AND_CHECKED",
        "candidate_id": candidate,
        "question_sha256": question_sha,
        "provider_receipt": provider_receipt,
        "pool50_scores_verified": 50,
        "provider_control_wrong_hybrid_order_rejected": True,
        "product_runtime_binding_verified": False,
        "product_regression_tests_executed": 0,
        "new_api_calls": 0,
        "qdrant_queries_attempted": 0,
        "new_local_pairs_scored": 0,
        "done_item_count": 7,
        "open_item_count": 5,
        "newly_completed_items_from_previous_open_set": [],
        "historical_gate": "FAIL",
        "selected_version_gate": "BLOCKED",
        "next_product_dependency": "BIND_HYBRID_AND_FINAL_RANK_PROVIDERS_TO_REAL_PIPELINE_CALL",
    }
    result["output"] = save("C135.R11_versioned_pool50_provider", result)
    print(json.dumps(result, ensure_ascii=False, indent=2))
    return result, pool50_provider


V3_C135, C135_POOL50_PROVIDER = run_v3_c135()
````

````python
# C138 — R11.3 — Local ranking inside the constructed opt-in product call
def run_v3_c138():
    def require(ok, code):
        if not ok:
            raise ValueError(code)

    def sha(raw):
        return hashlib.sha256(raw).hexdigest()

    reports = Path("reports/phase2/quality_improvement/NB22E.lab.v3")
    cwd = Path.cwd()
    root = next((p for p in (cwd, *cwd.parents) if (p / reports).is_dir()), None)
    require(root is not None, "C138_PROJECT_ROOT_NOT_FOUND")
    root = root.resolve()

    def bound(path, digest):
        path = (root / path).resolve()
        require(path.is_relative_to(root), "C138_PATH_OUTSIDE_PROJECT")
        raw = path.read_bytes()
        require(sha(raw) == digest, f"C138_HASH_CONFLICT:{path.name}")
        return raw

    def report(stem, digest):
        return json.loads(bound(reports / f"{stem}.{digest}.json", digest))

    c137_sha = "8a7a2fb1a0bbacb40b0316cd66de1798d4596fdfd517e6a889b5113880b9fe1b"
    c122_sha = "85492d51e95b2aa7dcb2a1a5f5cdb44a0bd5635550d736e38bb20dd5af8d209e"
    startup = report("C137.R11_constructed_product_instance_fixed_input", c137_sha)
    chain = report("C122.R11_development_rank_chain", c122_sha)
    require(
        startup["status"] == "CONSTRUCTED_PRODUCT_INSTANCE_FIXED_INPUT_TEST_PASSED"
        and chain["status"] == "QUOTA100_AND_CHAR_RRF_FULL_ORDERS_MATCHED"
        and (startup["candidate_id"], startup["question_sha256"])
        == (chain["candidate_id"], chain["question_sha256"]),
        "C138_PRIOR_EVIDENCE_CONFLICT",
    )

    history = globals().get("In")
    require(isinstance(history, list), "C138_IPYTHON_HISTORY_REQUIRED")

    def definition(name):
        header = f"def {name}():"
        entry = next(
            (item for item in reversed(history)
             if isinstance(item, str) and header in item.splitlines()),
            None,
        )
        require(entry is not None, f"C138_SOURCE_REQUIRED:{name}")
        nodes = [
            node for node in ast.parse(entry).body
            if isinstance(node, ast.FunctionDef) and node.name == name
        ]
        require(len(nodes) == 1, f"C138_SOURCE_AMBIGUOUS:{name}")
        return ast.get_source_segment(entry, nodes[0])

    def replace_once(source, old, new):
        require(source.count(old) == 1, f"C138_SOURCE_CONTRACT_CHANGED:{old}")
        return source.replace(old, new, 1)

    metrics = {
        "qdrant_queries_attempted": 0,
        "new_local_pairs_scored": 0,
        "char_channel_depth": 0,
    }

    def quiet_run(name, **overrides):
        function = globals().get(name)
        require(callable(function), f"C138_FUNCTION_REQUIRED:{name}")
        scope = dict(function.__globals__)
        scope.update(overrides)
        scope["print"] = lambda *args, **kwargs: None
        return type(function)(function.__code__, scope)()

    def live_hybrid(pipeline, question, candidate_id, reference):
        require(metrics["qdrant_queries_attempted"] == 0, "C138_DUPLICATE_HYBRID_CALL")
        client = pipeline.retriever.client
        expected_path = (root / startup["dense_storage_path"]).resolve()

        class BorrowedClient:
            def __getattr__(self, name):
                return getattr(client, name)

            def close(self):
                # The constructed pipeline owns this client.
                pass

        def borrow(*, path):
            require(Path(path).resolve() == expected_path, "C138_DENSE_STORAGE_CONFLICT")
            return BorrowedClient()

        observed = quiet_run("run_v3_c126", QdrantClient=borrow)
        metrics["qdrant_queries_attempted"] = observed["qdrant_queries_attempted"]
        require(
            observed["status"] == "LOCAL_HYBRID100_ORDER_MATCHED"
            and observed["observed_hybrid100_order"] == reference
            and observed["candidate_id"] == candidate_id
            and observed["question_sha256"] == sha(question.encode("utf-8"))
            and observed["dense_collection"] == startup["dense_collection"]
            and observed["qdrant_queries_attempted"] == 1
            and observed["new_api_calls"] == 0,
            f"C138_HYBRID100_CONFLICT:{observed.get('blocker')}",
        )
        metrics["hybrid100_probe_receipt"] = observed["output"]
        return {
            "candidate_id": candidate_id,
            "question_sha256": observed["question_sha256"],
            "ordered_chunk_ids": observed["observed_hybrid100_order"],
            "receipt_sha256": observed["output"]["file_bytes_sha256"],
        }

    def live_rank(question, candidate_id, hybrid100, pool50_order, pool50_scores):
        require(metrics["new_local_pairs_scored"] == 0, "C138_DUPLICATE_LOCAL100_CALL")
        scored = quiet_run("run_v3_c120")
        metrics["new_local_pairs_scored"] = scored["new_local_pairs_scored"]
        require(
            scored["status"] == "ONE_DEVELOPMENT_LOCAL100_INPUT_BUILT"
            and scored["candidate_id"] == candidate_id
            and scored["question_sha256"] == sha(question.encode("utf-8"))
            and scored["hybrid100_order"] == list(hybrid100)
            and scored["scored_pool50_order"] == list(pool50_order)
            and all(scored["scores_by_chunk_id"][cid] == score
                    for cid, score in pool50_scores.items())
            and scored["new_local_pairs_scored"] == 50
            and scored["new_api_calls"] == 0,
            f"C138_LOCAL100_CONFLICT:{scored.get('blocker')}",
        )
        metrics["local100_probe_receipt"] = scored["output"]

        notebook_path = "notebooks/phase2/22d_full_corpus_retrieval_expansion.ipynb"
        notebook_sha = "fff5bd072a4977820882045451b61d66f8621ef9a4918775e4534800f0ae1480"
        notebook = json.loads(bound(notebook_path, notebook_sha))
        names = {
            "normalized_words", "char_tokens", "build_index", "lexical_search",
            "lexical_fusion",
        }
        nodes = [
            node for node in ast.parse("".join(notebook["cells"][12]["source"])).body
            if isinstance(node, ast.FunctionDef) and node.name in names
        ]
        require(len(nodes) == len(names) and {node.name for node in nodes} == names,
                "C138_NB22D_DEFINITIONS_REQUIRED")
        corpus_path = (
            "data/phase2/chunks/nb11_benchmark_v1/baseline_phase1_char_safe_v1.jsonl"
        )
        chunks = {}
        for line in bound(corpus_path, scored["input_file_hashes"]["chunks"]).splitlines():
            if line.strip():
                row = json.loads(line)
                cid = row["chunk_id"]
                require(cid not in chunks, "C138_DUPLICATE_CHUNK")
                chunks[cid] = row
        require(len(chunks) == 3488, "C138_CORPUS_COUNT_CONFLICT")
        scope = dict(globals())
        scope.update(require=require, CONFIG={
            "char_max_df_fraction": 0.30,
            "bm25": {"k1": 1.2, "b": 0.3},
            "rrf_k": 10,
            "incumbent_rank_weight": 0.75,
        })
        exec(compile(ast.Module(body=nodes, type_ignores=[]), notebook_path, "exec"), scope)
        index = scope["build_index"](chunks, "char")
        char_order = scope["lexical_search"](index, question, "char")
        channel_ref = chain["source_rows"]["NB22D_A03"]
        channel_sha = chain["input_file_hashes"]["NB22D_A03"]
        channel_row = json.loads(bound(
            channel_ref["path"], channel_sha,
        ).splitlines()[channel_ref["line_number"] - 1])
        require(channel_row["candidate_id"] == candidate_id
                and char_order == channel_row["channels"]["char"],
                "C138_CHAR_CHANNEL_ORDER_CONFLICT")
        metrics["char_channel_depth"] = len(char_order)

        hybrid_rank = {cid: i for i, cid in enumerate(hybrid100, 1)}
        local_rank = {cid: i for i, cid in enumerate(scored["local100_order"], 1)}
        require(set(hybrid_rank) == set(local_rank), "C138_LOCAL100_POOL_CONFLICT")
        ranked = sorted(hybrid100, key=lambda cid: (
            -(0.75 / (10 + local_rank[cid]) + 0.25 / (10 + hybrid_rank[cid])), cid,
        ))
        pool = set(pool50_order)
        top = []
        newcomer_used = False
        for cid in ranked:
            if cid not in pool:
                if newcomer_used:
                    continue
                newcomer_used = True
            top.append(cid)
            if len(top) == 10:
                break
        chosen = set(top)
        quota_order = top + [cid for cid in ranked if cid not in chosen]
        require(quota_order == chain["quota100_order"], "C138_QUOTA100_ORDER_CONFLICT")
        fused = scope["lexical_fusion"](
            quota_order, {"char": char_order}, {"channels": ["char"]},
        )
        require(fused == chain["char_rrf_order"], "C138_CHAR_RRF_ORDER_CONFLICT")
        return fused[:10]

    # Reuse the successful C137 constructor and C136 receipt checks.
    template = definition("run_v3_c136")
    template = replace_once(
        template,
        '        events.append("hybrid100_receipt_bound")\n'
        '        return {\n'
        '            "candidate_id": candidate,\n'
        '            "question_sha256": question_hash,\n'
        '            "ordered_chunk_ids": list(hybrid_ids),\n'
        '            "receipt_sha256": hybrid_ref["file_bytes_sha256"],\n'
        '        }',
        '        supplied = c138_live_hybrid(pipeline, question, candidate_id, hybrid_ids)\n'
        '        events.append("hybrid100_receipt_bound")\n'
        '        return supplied',
    )
    template = replace_once(
        template,
        '        events.append("verified_final_order_supplied")',
        '        fresh_top = c138_live_rank(\n'
        '            question, candidate_id, hybrid100, pool50_order, pool50_scores,\n'
        '        )\n'
        '        require(fresh_top == top10, "C138_FINAL_TOP10_CONFLICT")\n'
        '        chunks_by_id = {chunk["chunk_id"]: chunk for chunk in chunks}\n'
        '        events.append("verified_final_order_supplied")',
    )
    template = replace_once(
        template, '            "ordered_chunks": chunks,',
        '            "ordered_chunks": [chunks_by_id[cid] for cid in fresh_top],',
    )
    template = replace_once(
        template,
        '                "c135_file_bytes_sha256": c135_sha,',
        '                "c135_file_bytes_sha256": c135_sha,\n'
        '                "c122_file_bytes_sha256": c138_c122_sha,\n'
        '                "local100_probe_receipt_sha256": '
        'c138_metrics["local100_probe_receipt"]["file_bytes_sha256"],',
    )
    template = replace_once(
        template,
        '                and identity["hybrid100_provider_receipt_sha256"] == hybrid_ref["file_bytes_sha256"],',
        '                and identity["hybrid100_provider_receipt_sha256"]\n'
        '                == c138_metrics["hybrid100_probe_receipt"]["file_bytes_sha256"],',
    )
    template = replace_once(
        template,
        '        "qdrant_queries_attempted": 0,\n        "new_local_pairs_scored": 0,',
        '        **c138_metrics,\n'
        '        "rank_providers_executed_in_constructed_product_call": True,',
    )
    template = replace_once(
        template, '    result["output"] = save(',
        '    result["input_file_hashes"].update(C137=c138_c137_sha, C122=c138_c122_sha)\n'
        '    result["output"] = save(',
    )

    adapter = definition("run_v3_c137")
    for old, new in (
        ("CONSTRUCTED_PRODUCT_INSTANCE_FIXED_INPUT_TEST_PASSED",
         "CONSTRUCTED_PRODUCT_INSTANCE_LOCAL_RANK_CHAIN_PASSED"),
        ("FIXED_INPUT_CONSTRUCTED_PRODUCT_INSTANCE_STUB_GENERATOR",
         "LOCAL_RANK_CHAIN_CONSTRUCTED_PRODUCT_INSTANCE_STUB_GENERATOR"),
        ("RUN_VERSIONED_RANK_PROVIDERS_INSIDE_CONFIGURED_PRODUCT_CALL",
         "ASSESS_REMAINING_SELECTED_RUNTIME_REGRESSION_PREREQUISITES"),
        ("C137.R11_constructed_product_instance_fixed_input",
         "C138.R11_constructed_product_instance_local_rank_chain"),
        ('\'"cell": "C137"\'', '\'"cell": "C138"\''),
        ("C137.R11_product_", "C138.R11_product_"),
    ):
        adapter = replace_once(adapter, old, new)

    scope = dict(globals())
    scope.update(
        In=[template],
        c138_live_hybrid=live_hybrid,
        c138_live_rank=live_rank,
        c138_metrics=metrics,
        c138_c137_sha=c137_sha,
        c138_c122_sha=c122_sha,
    )
    exec(compile(adapter, "<C138:adapted-C137>", "exec"), scope)
    return scope["run_v3_c137"]()


V3_C138 = run_v3_c138()
````

Hybrid100 ترتیب و رسید را می‌دهد، Pool50 امتیازهای متصل به همان سؤال و مجموعه را تأمین می‌کند، و رتبه‌بند ترتیب نهایی را می‌سازد. ثبت رسید ورودی و خروجی در متد محصول انجام می‌شود.

````python
# C139 — R11.3 — Persist the explicit route after a differential contract check
def run_v3_c139():
    def require(ok, code):
        if not ok:
            raise ValueError(code)

    def sha(raw):
        return hashlib.sha256(raw).hexdigest()

    reports = Path("reports/phase2/quality_improvement/NB22E.lab.v3")
    cwd = Path.cwd()
    root = next((p for p in (cwd, *cwd.parents) if (p / reports).is_dir()), None)
    require(root is not None, "C139_PROJECT_ROOT_NOT_FOUND")
    root = root.resolve()

    def project_path(relative):
        relative = Path(relative)
        path = (root / relative).resolve()
        require(not relative.is_absolute() and path.is_relative_to(root),
                "C139_PATH_OUTSIDE_PROJECT")
        return path

    def read_bound(relative, digest):
        raw = project_path(relative).read_bytes()
        require(sha(raw) == digest, f"C139_HASH_CONFLICT:{relative}")
        return raw

    def write_once(relative, raw):
        path = project_path(relative)
        try:
            with path.open("xb") as handle:
                handle.write(raw)
                handle.flush()
                os.fsync(handle.fileno())
            status = "CREATED"
        except FileExistsError:
            status = "ALREADY_PRESENT"
        require(path.read_bytes() == raw, f"C139_EXISTING_BYTES_CONFLICT:{relative}")
        return status

    c138_sha = "1df8a9e4c0fcc6a4bfcce4edb4356f2557f71e312aed48f501b89a8c5464e910"
    prior = json.loads(read_bound(
        reports / f"C138.R11_constructed_product_instance_local_rank_chain.{c138_sha}.json",
        c138_sha,
    ))
    require(
        prior["status"] == "CONSTRUCTED_PRODUCT_INSTANCE_LOCAL_RANK_CHAIN_PASSED"
        and prior["rank_providers_executed_in_constructed_product_call"] is True,
        "C139_SUCCESSFUL_C138_REQUIRED",
    )
    pipeline_sha = prior["input_file_hashes"]["pipeline_source"]
    read_bound("src/pipeline/rag_pipeline.py", pipeline_sha)

    name = "ask_product_char_rrf"
    active = globals().get(name)
    history = globals().get("In")
    source_index = 43
    require(
        callable(active) and isinstance(history, list) and len(history) > source_index,
        "C139_ACTIVE_C134_AND_INPUT_43_REQUIRED",
    )

    entry = history[source_index]
    definitions = [
        node for node in ast.parse(entry).body
        if isinstance(node, ast.FunctionDef) and node.name == name
    ]
    require(len(definitions) == 1 and not definitions[0].decorator_list,
            "C139_EXPECTED_ONE_PLAIN_C134_DEFINITION")
    caller_source = ast.get_source_segment(entry, definitions[0])
    require(bool(caller_source), "C139_C134_SOURCE_EXTRACTION_FAILED")

    relative = Path("src/pipeline/char_rrf_route.py")
    module_source = (
        '"""Explicit product entry for the opt-in char_rrf route."""\n\n'
        "import hashlib\n"
        "import math\n\n\n"
        + caller_source + "\n"
    )
    namespace = {}
    exec(compile(module_source, relative.as_posix(), "exec", dont_inherit=True), namespace)
    exported = namespace[name]
    require(
        exported.__defaults__ == active.__defaults__
        and exported.__kwdefaults__ == active.__kwdefaults__
        and exported.__annotations__ == active.__annotations__,
        "C139_EXPORTED_CALLER_DEFAULTS_OR_ANNOTATIONS_CONFLICT",
    )

    def exercise(route, scenario):
        question = "سؤال آزمون اتصال"
        candidate = "c139-contract-probe"
        question_hash = sha(question.encode("utf-8"))
        hybrid_ids = [f"chunk-{i:03d}" for i in range(100)]
        pool_ids = list(reversed(hybrid_ids[:50]))
        scores = {cid: float(i) for i, cid in enumerate(pool_ids)}
        calls = []

        def hybrid_provider(**kwargs):
            calls.append(("hybrid", kwargs))
            return {
                "candidate_id": candidate,
                "question_sha256": "0" * 64 if scenario == "wrong_question" else question_hash,
                "ordered_chunk_ids": hybrid_ids,
                "receipt_sha256": "1" * 64,
            }

        def pool_provider(**kwargs):
            calls.append(("pool", kwargs))
            order = pool_ids if scenario != "wrong_pool" else pool_ids[:-1] + [hybrid_ids[50]]
            return {
                "candidate_id": candidate,
                "question_sha256": question_hash,
                "ordered_chunk_ids": order,
                "scores_by_chunk_id": scores,
                "receipt_sha256": "2" * 64,
            }

        def rank_provider(**kwargs):
            calls.append(("rank", kwargs))
            return {
                "candidate_id": candidate,
                "question_sha256": question_hash,
                "selected_variant": "char_rrf",
                "runtime_identity": {"ranker": "contract-probe"},
                "ordered_chunks": [{"chunk_id": pool_ids[0], "text": "متن آزمون"}],
            }

        def write_receipt(role, payload):
            calls.append(("write_receipt", role, payload))
            return "3" * 64

        class PipelineProbe:
            def ask_with_char_rrf(
                self, question, *, candidate_id, rank_sources, write_receipt, top_k
            ):
                calls.append(("pipeline", question, candidate_id, top_k))
                ranked = rank_sources(
                    question=question, candidate_id=candidate_id, top_k=top_k,
                )
                return {"ranked": ranked, "receipt_sha256": write_receipt("input", ranked)}

        try:
            result = route(
                PipelineProbe(), question=question, candidate_id=candidate,
                hybrid100_provider=hybrid_provider, pool50_provider=pool_provider,
                rank_provider=rank_provider, write_receipt=write_receipt,
            )
        except ValueError as error:
            return {"error": str(error), "calls": calls}
        return {"result": result, "calls": calls}

    checks = {}
    scenarios = {
        "valid": ["pipeline", "hybrid", "pool", "rank", "write_receipt"],
        "wrong_pool": ["pipeline", "hybrid", "pool"],
        "wrong_question": ["pipeline", "hybrid"],
    }
    for scenario, expected_calls in scenarios.items():
        previous = exercise(active, scenario)
        current = exercise(exported, scenario)
        require(current == previous, f"C139_BEHAVIOR_DIFFERS:{scenario}")
        require([call[0] for call in current["calls"]] == expected_calls,
                f"C139_UNEXPECTED_CALL_ORDER:{scenario}")
        if scenario == "valid":
            identity = current["result"]["ranked"]["runtime_identity"]
            require(
                identity["hybrid100_provider_receipt_sha256"] == "1" * 64
                and identity["pool50_provider_receipt_sha256"] == "2" * 64,
                "C139_PROVIDER_RECEIPT_BINDING_CONFLICT",
            )
        else:
            require("error" in current, f"C139_INVALID_INPUT_ACCEPTED:{scenario}")
        checks[scenario] = "PASS"

    module_bytes = module_source.encode("utf-8")
    write_status = write_once(relative, module_bytes)
    result = {
        "cell": "C139",
        "roadmap_action": "R11",
        "roadmap_subtask": "R11.3",
        "status": "OPT_IN_CALLER_PERSISTED_IN_PRODUCT_SOURCE",
        "input_file_hashes": {"C138": c138_sha, "pipeline_source": pipeline_sha},
        "module": {
            "path": relative.as_posix(),
            "file_bytes_sha256": sha(module_bytes),
            "write_status": write_status,
            "export": name,
        },
        "caller_validation": {
            "source_history_input": source_index,
            "source_definition_sha256": sha(caller_source.encode("utf-8")),
            "method": "DIFFERENTIAL_CONTRACT_CHECK_ON_THREE_SCENARIOS",
            "checks": checks,
            "bytecode_equivalence_verified": False,
        },
        "product_runtime_binding_verified": False,
        "product_regression_tests_executed": 0,
        "qdrant_queries_attempted": 0,
        "new_local_pairs_scored": 0,
        "new_api_calls": 0,
        "done_item_count": 7,
        "open_item_count": 5,
        "selected_version_gate": "BLOCKED",
        "next_product_dependency": "BIND_VERSIONED_PROVIDERS_AT_CONFIGURED_STARTUP",
    }
    payload = json.dumps(
        result, ensure_ascii=False, sort_keys=True, indent=2, allow_nan=False,
    ).encode("utf-8")
    digest = sha(payload)
    output = reports / f"C139.R11_persisted_opt_in_caller.{digest}.json"
    write_once(output, payload)
    result["output"] = {
        "path": output.as_posix(),
        "file_bytes_sha256": digest,
        "reload_validation": "PASS",
    }
    print(json.dumps(result, ensure_ascii=False, indent=2))
    return result


V3_C139 = run_v3_c139()
````

````python
# C142 — R11.3 — Bind standalone execution evidence to the regression population
def run_v3_c142():
    def require(ok, code):
        if not ok:
            raise ValueError(code)

    def sha(raw):
        return hashlib.sha256(raw).hexdigest()

    reports = Path("reports/phase2/quality_improvement/NB22E.lab.v3")
    cwd = Path.cwd()
    root = next((p for p in (cwd, *cwd.parents) if (p / reports).is_dir()), None)
    require(root is not None, "C142_PROJECT_ROOT_NOT_FOUND")
    root = root.resolve()

    def load(relative, digest):
        relative = Path(relative)
        path = (root / relative).resolve()
        require(not relative.is_absolute() and path.is_relative_to(root),
                "C142_PATH_OUTSIDE_PROJECT")
        raw = path.read_bytes()
        require(sha(raw) == digest, f"C142_HASH_CONFLICT:{relative}")
        return json.loads(raw)

    pins = {
        "C080": (
            "C080.R11_diagnostic_residuals_and_execution_requirements",
            "15dba7dddba32ec40bd63547637735794932a72c02a970df4e45b1f49678857b",
        ),
        "C090": (
            "C090.R11_fixed_input_diagnostic_receipts",
            "180520de851dae9fc9265574d36805e60f57e94ae357936db8601969dcd91a47",
        ),
        "C141": (
            "C141.R11_standalone_local_rank_chain",
            "8d45d6d11d8b5fe31863c585b4b5b354265562e47647eb094db415f45e1ca418",
        ),
    }
    documents = {
        key: load(reports / f"{stem}.{digest}.json", digest)
        for key, (stem, digest) in pins.items()
    }
    requirements = documents["C080"]
    targets = documents["C090"]
    run = documents["C141"]

    require(
        requirements["status"] == "DIAGNOSTIC_RESIDUALS_AND_PRODUCT_EXECUTION_REQUIREMENTS_RECORDED"
        and targets["status"] == "FIXED_INPUT_DIAGNOSTIC_RECEIPTS_RECONSTRUCTED"
        and targets["input_file_hashes"]["C080"] == pins["C080"][1]
        and run["status"] == "STANDALONE_LOCAL_RANK_CHAIN_PASSED"
        and run["rank_providers_executed_in_standalone_product_call"] is True,
        "C142_PRIOR_EVIDENCE_CONFLICT",
    )

    receipt_refs = run["receipts"]
    input_ref = receipt_refs["input"]
    output_ref = receipt_refs["output"]
    input_receipt = load(input_ref["path"], input_ref["file_bytes_sha256"])
    output_receipt = load(output_ref["path"], output_ref["file_bytes_sha256"])
    require(
        input_receipt["candidate_id"] == run["candidate_id"]
        and input_receipt["question_sha256"] == run["question_sha256"]
        and input_receipt["selected_variant"] == "char_rrf"
        and output_receipt["input_receipt_file_bytes_sha256"] == input_ref["file_bytes_sha256"]
        and run["events"][-3:] == [
            "input_receipt_written",
            "stub_verified_persisted_input",
            "output_receipt_written",
        ],
        "C142_C141_RECEIPT_CHAIN_CONFLICT",
    )

    matched = targets["c036_order_matched_candidate_ids"]
    unverified = targets["rank_origin_unverified_candidate_ids"]
    matched_ids, unverified_ids = set(matched), set(unverified)
    target_ids = matched_ids | unverified_ids
    require(
        len(matched) == len(matched_ids) == targets["c036_order_match_count"] == 18
        and len(unverified) == len(unverified_ids) == targets["rank_origin_unverified_count"] == 16
        and matched_ids.isdisjoint(unverified_ids)
        and len(target_ids) == targets["candidate_receipt_count"] == 34,
        "C142_TARGET_POPULATION_CONFLICT",
    )

    scenario_counts = targets["scenario_counts"]
    required = requirements["product_execution_requirement_counts"]
    require(
        scenario_counts == {"ANSWER_DECISION": 31, "CITATION_WITNESS": 5}
        and required["SELECTED_VARIANT_AND_RUNTIME_RECEIPT"] == 36
        and required["SELECTED_RESPONSE_DOCUMENT_AND_BYTE_HASH"] == 36
        and required["EXPECTED_BEHAVIOR_AND_REQUIRED_LIMIT_ASSESSED"] == 31
        and required["EXACT_SELECTED_CLAIM_LINK_AND_QUALIFIED_WITNESS_OR_EXPLICIT_VERDICT"] == 5,
        "C142_SCENARIO_REQUIREMENT_CONFLICT",
    )

    candidate = run["candidate_id"]
    rows = [
        {
            "candidate_id": cid,
            "diagnostic_order_status": (
                "C036_FULL_ORDER_MATCH_RECORDED" if cid in matched_ids
                else "RANK_ORIGIN_UNVERIFIED_RECONSTRUCTION_DEFERRED"
            ),
            "candidate_id_matches_c141": cid == candidate,
            "selected_runtime_receipt_established_by_c141": False,
            "selected_response_or_citation_witness_established_by_c141": False,
        }
        for cid in sorted(target_ids)
    ]

    result = {
        "cell": "C142",
        "roadmap_action": "R11",
        "roadmap_subtask": "R11.3",
        "status": "STANDALONE_PROBE_AND_REGRESSION_TARGETS_RECONCILED",
        "scope": "C080_C090_C141_BOUND_EVIDENCE",
        "input_file_hashes": {
            **{key: digest for key, (_, digest) in pins.items()},
            "C141_input_receipt": input_ref["file_bytes_sha256"],
            "C141_output_receipt": output_ref["file_bytes_sha256"],
        },
        "c141_candidate_id": candidate,
        "c141_candidate_in_regression_population": candidate in target_ids,
        "c141_generator_kind": "OFFLINE_STUB",
        "target_candidate_count": len(rows),
        "scenario_counts": scenario_counts,
        "diagnostic_order_matches": len(matched_ids),
        "diagnostic_rank_origin_unverified": len(unverified_ids),
        "selected_runtime_receipts_established_by_c141": 0,
        "selected_response_or_witness_scenarios_established_by_c141": 0,
        "target_candidate_rows": rows,
        "pending_diagnostic_answer_spec_ids": requirements["pending_regression_spec_ids"],
        "rank_reconstruction_for_unverified_candidates": "DEFERRED",
        "product_regression_tests_executed": 0,
        "new_api_calls": 0,
        "qdrant_queries_attempted": 0,
        "new_local_pairs_scored": 0,
        "done_item_count": 7,
        "open_item_count": 5,
        "newly_completed_items_from_previous_open_set": [],
        "selected_version_gate": "BLOCKED",
        "next_product_dependency": (
            "BIND_ONE_TARGET_SCENARIO_TO_SELECTED_RUNTIME_INPUT_AND_EXPECTED_BEHAVIOR"
        ),
    }

    raw = json.dumps(
        result, ensure_ascii=False, sort_keys=True, indent=2, allow_nan=False,
    ).encode("utf-8")
    digest = sha(raw)
    relative = reports / f"C142.R11_standalone_probe_target_reconciliation.{digest}.json"
    path = root / relative
    try:
        with path.open("xb") as handle:
            handle.write(raw)
            handle.flush()
            os.fsync(handle.fileno())
    except FileExistsError:
        pass
    require(path.read_bytes() == raw, "C142_REPORT_RELOAD_CONFLICT")

    result["output"] = {
        "path": relative.as_posix(),
        "file_bytes_sha256": digest,
        "reload_validation": "PASS",
    }
    print(json.dumps(
        {key: value for key, value in result.items() if key != "target_candidate_rows"},
        ensure_ascii=False, indent=2,
    ))
    return result


V3_C142 = run_v3_c142()
````

### سل ۱۴۳ — اتصال یک سناریوی هدف به ورودی تشخیصی و انتظار آزمون

رسیدهای C142، C037 و C043 با هش کنترل می‌شوند. یک سناریوی تصمیم پاسخ
از پرونده‌های دارای تطابق ترتیب تشخیصی انتخاب می‌شود. اتصال فقط بر اساس
شناسهٔ صریح سناریو و پرونده انجام می‌شود. متن سؤال و انتظار ثبت‌شده از
رکوردهای همان سناریو خوانده می‌شوند.

ترتیب تشخیصی و انتظار آزمون جدا از رسید اجرای نسخهٔ منتخب ثبت می‌شوند.
هیچ رتبه، پاسخ یا رأی تازه‌ای تولید نمی‌شود.

</div>

````python
# C143 — R11.3 — Bind one target answer scenario to existing evidence
def run_v3_c143():
    reports = Path("reports/phase2/quality_improvement/NB22E.lab.v3")
    cwd = Path.cwd()
    root = next((p for p in (cwd, *cwd.parents) if (p / reports).is_dir()), None)
    if root is None:
        raise ValueError("C143_PROJECT_ROOT_NOT_FOUND")

    def sha(raw):
        return hashlib.sha256(raw).hexdigest()

    def load(stem, digest):
        raw = (root / reports / f"{stem}.{digest}.json").read_bytes()
        if sha(raw) != digest:
            raise ValueError(f"C143_HASH_CONFLICT:{stem}")
        return json.loads(raw)

    pins = {
        "C142": (
            "C142.R11_standalone_probe_target_reconciliation",
            "166c44ef3935c3b8f5fde92e4166290c38e7391bddaa038585151276845cb04e",
        ),
        "C037": (
            "C037.R11_response_regression_inputs",
            "249ab2cb0171bbf79c453ee9bf0310c4319866c41a65008a7d46d966c96fa255",
        ),
        "C043": (
            "C043.R11_scenario_review_inputs",
            "811af55fabd63d695bd2767c762055c4997d81e4acd7c36f7bdde717421e6145",
        ),
    }
    c142, c037, c043 = (load(*pins[cell]) for cell in ("C142", "C037", "C043"))
    if c142["status"] != "STANDALONE_PROBE_AND_REGRESSION_TARGETS_RECONCILED":
        raise ValueError("C143_SUCCESSFUL_C142_REQUIRED")

    eligible = {
        row["candidate_id"]
        for row in c142["target_candidate_rows"]
        if row["diagnostic_order_status"] == "C036_FULL_ORDER_MATCH_RECORDED"
    }
    if len(eligible) != 18:
        raise ValueError("C143_DIAGNOSTIC_POPULATION_CONFLICT")

    def records(value, pointer=""):
        if isinstance(value, dict):
            yield pointer, value
            for key, child in value.items():
                token = key.replace("~", "~0").replace("/", "~1")
                yield from records(child, f"{pointer}/{token}")
        elif isinstance(value, list):
            for index, child in enumerate(value):
                yield from records(child, f"{pointer}/{index}")

    def scenario_id(row):
        ids = {
            row[key] for key in ("regression_spec_id", "regression_id", "spec_id")
            if isinstance(row.get(key), str) and row[key].startswith("REG_")
        }
        return next(iter(ids)) if len(ids) == 1 else None

    def is_answer_decision(row):
        return any(
            row.get(key) == "ANSWER_DECISION"
            for key in ("scenario_type", "scenario_kind", "kind", "type", "scenario_family")
        )

    specs = {}
    reviews = {}
    for label, document, index in (("C037", c037, specs), ("C043", c043, reviews)):
        for pointer, row in records(document):
            cid = row.get("candidate_id")
            sid = scenario_id(row)
            if isinstance(cid, str) and cid in eligible and sid:
                index.setdefault((cid, sid), []).append((pointer, row))

    answer_keys = sorted(
        key for key in specs.keys() & reviews.keys()
        if any(is_answer_decision(row) for _, row in specs[key] + reviews[key])
    )
    expectation_keys = (
        "expected_behavior", "expected_decision", "expected_answer_behavior",
        "required_limit", "required_limits", "expected_behavior_and_required_limit",
        "expected_behavior_fa", "required_limit_fa",
    )

    def direct_fields(entries, keys):
        return [
            {"pointer": f"{pointer}/{key}", "value": row[key]}
            for pointer, row in entries
            for key in keys
            if row.get(key) not in (None, "", [], {})
        ]

    selected = None
    unresolved = []
    for candidate, sid in answer_keys:
        spec_rows = specs[(candidate, sid)]
        review_rows = reviews[(candidate, sid)]
        expected = direct_fields(spec_rows, expectation_keys)
        questions = [
            entry for entry in direct_fields(review_rows, ("question", "question_text"))
            if isinstance(entry["value"], str) and entry["value"].strip()
        ]
        distinct_questions = {entry["value"] for entry in questions}
        if not expected or len(distinct_questions) != 1:
            unresolved.append({
                "candidate_id": candidate,
                "regression_spec_id": sid,
                "expectation_fields_found": len(expected),
                "distinct_question_texts_found": len(distinct_questions),
            })
            continue

        question = questions[0]["value"]
        selected = {
            "candidate_id": candidate,
            "regression_spec_id": sid,
            "scenario_type": "ANSWER_DECISION",
            "question": question,
            "question_sha256": sha(question.encode("utf-8")),
            "question_sources_in_c043": [entry["pointer"] for entry in questions],
            "expectation_fields_in_c037": expected,
            "c037_scenario_pointers": [pointer for pointer, _ in spec_rows],
            "c043_scenario_pointers": [pointer for pointer, _ in review_rows],
            "diagnostic_order_status": "C036_FULL_ORDER_MATCH_RECORDED",
        }
        break

    result = {
        "cell": "C143",
        "roadmap_action": "R11",
        "roadmap_subtask": "R11.3",
        "status": (
            "ONE_TARGET_SCENARIO_BOUND_TO_DIAGNOSTIC_EVIDENCE"
            if selected else "TARGET_SCENARIO_BINDING_BLOCKED"
        ),
        "input_file_hashes": {key: digest for key, (_, digest) in pins.items()},
        "selection_rule": "FIRST_SORTED_ANSWER_SCENARIO_WITH_EXPLICIT_BINDINGS",
        "eligible_diagnostic_candidates": len(eligible),
        "c037_explicit_scenario_candidate_keys": len(specs),
        "c043_explicit_scenario_candidate_keys": len(reviews),
        "matched_answer_scenario_keys": len(answer_keys),
        "selected_scenario": selected,
        "incomplete_bindings_encountered": unresolved,
        "selected_runtime_receipt_verified": False,
        "product_regression_tests_executed": 0,
        "new_api_calls": 0,
        "target_case_rankings_reconstructed": 0,
        "rank_reconstruction_for_unverified_candidates": "DEFERRED",
        "done_item_count": 7,
        "open_item_count": 5,
        "newly_completed_items_from_previous_open_set": [],
        "selected_version_gate": "BLOCKED",
        "next_product_dependency": (
            "VERIFY_SELECTED_RUNTIME_INPUT_RECEIPT_FOR_BOUND_SCENARIO"
            if selected else "RESOLVE_EXACT_SCENARIO_FIELD_PATHS_IN_C037_AND_C043"
        ),
    }

    raw = json.dumps(
        result, ensure_ascii=False, sort_keys=True, indent=2, allow_nan=False
    ).encode("utf-8")
    digest = sha(raw)
    relative = reports / f"C143.R11_target_scenario_binding.{digest}.json"
    path = root / relative
    try:
        with path.open("xb") as handle:
            handle.write(raw)
            handle.flush()
            os.fsync(handle.fileno())
    except FileExistsError:
        pass
    if path.read_bytes() != raw:
        raise ValueError("C143_REPORT_RELOAD_CONFLICT")

    result["output"] = {
        "path": relative.as_posix(),
        "file_bytes_sha256": digest,
        "reload_validation": "PASS",
    }
    print(json.dumps(result, ensure_ascii=False, indent=2))
    return result


V3_C143 = run_v3_c143()
````

### سل ۱۴۴ — اتصال سناریوی هدف به تعریف اصلی آزمون

شناسهٔ سناریو و پرونده از تطبیق C143 مصرف می‌شوند. متن سؤال از Golden
نسخه‌بندی‌شده و تعریف کامل آزمون از C078 خوانده می‌شوند. هش فایل‌ها و
هویت ردیف‌ها کنترل و مسیر دقیق هر شاهد ثبت می‌شود.

این سل ورودی و انتظار آزمون را آماده می‌کند. هیچ رتبه، پاسخ یا رأی تازه‌ای
تولید نمی‌شود و Gate نسخهٔ منتخب باز می‌ماند.

</div>

````python
# C144 — R11.3 — Bind one matched target to its original spec and Golden question
def run_v3_c144():
    def require(ok, code):
        if not ok:
            raise ValueError(code)

    def sha(raw):
        return hashlib.sha256(raw).hexdigest()

    reports = Path("reports/phase2/quality_improvement/NB22E.lab.v3")
    cwd = Path.cwd()
    root = next((p for p in (cwd, *cwd.parents) if (p / reports).is_dir()), None)
    require(root is not None, "C144_PROJECT_ROOT_NOT_FOUND")
    root = root.resolve()

    def read_bound(relative, digest):
        relative = Path(relative)
        path = (root / relative).resolve()
        require(
            not relative.is_absolute() and path.is_relative_to(root),
            "C144_PATH_OUTSIDE_PROJECT",
        )
        raw = path.read_bytes()
        require(sha(raw) == digest, f"C144_HASH_CONFLICT:{relative}")
        return raw

    c143_sha = "2e42c080f07d549e32a01cd4a6df0209f25fca0cf8e5ac63e1732fa0b1cb9c53"
    c142_sha = "166c44ef3935c3b8f5fde92e4166290c38e7391bddaa038585151276845cb04e"
    c078_sha = "09b05f51fbadd8a090b14c1c2ee65036987aa8e36555354535b983ebffe02006"
    golden_sha = "b9ca18e710686ddca48a25c637840b716adb306717cafd8ec7c96ce9e182e6f2"

    spec_id = "REG_cd77774a3651556cdb88c9e2"
    candidate = "golden_v2_candidate_4d16e7b7266a0c01d12a1036"
    spec_path = (
        Path("reports/phase2/evaluation/NB21C.C078.action42_final")
        / f"C078.regression_specs.{c078_sha}.json"
    )
    golden_path = Path("artifacts/phase2/golden_dataset_v2/golden_dataset_v2.jsonl")

    result = {
        "cell": "C144",
        "roadmap_action": "R11",
        "roadmap_subtask": "R11.3",
        "candidate_id": candidate,
        "regression_spec_id": spec_id,
        "scenario_type": "ANSWER_DECISION",
        "input_file_hashes": {
            "C143": c143_sha, "C142": c142_sha,
            "C078_original_specs": c078_sha, "golden": golden_sha,
        },
        "selected_runtime_receipt_verified": False,
        "product_regression_tests_executed": 0,
        "new_api_calls": 0,
        "target_case_rankings_reconstructed": 0,
        "rank_reconstruction_for_unverified_candidates": "DEFERRED",
        "done_item_count": 7,
        "open_item_count": 5,
        "newly_completed_items_from_previous_open_set": [],
        "selected_version_gate": "BLOCKED",
    }

    def walk(value, pointer=""):
        if isinstance(value, dict):
            yield pointer, value
            for key, child in value.items():
                token = key.replace("~", "~0").replace("/", "~1")
                yield from walk(child, f"{pointer}/{token}")
        elif isinstance(value, list):
            for index, child in enumerate(value):
                yield from walk(child, f"{pointer}/{index}")

    stage = "BIND_EXISTING_SCENARIO_MATCH"
    try:
        c143 = json.loads(read_bound(
            reports / f"C143.R11_target_scenario_binding.{c143_sha}.json", c143_sha,
        ))
        c142 = json.loads(read_bound(
            reports / f"C142.R11_standalone_probe_target_reconciliation.{c142_sha}.json",
            c142_sha,
        ))
        require(
            c143["input_file_hashes"]["C142"] == c142_sha
            and c143["matched_answer_scenario_keys"] == 18,
            "C144_C143_BINDING_CONFLICT",
        )
        matches = [
            row for row in c143["incomplete_bindings_encountered"]
            if (row["candidate_id"], row["regression_spec_id"]) == (candidate, spec_id)
        ]
        require(len(matches) == 1, "C144_EXPECTED_ONE_C143_SCENARIO_MATCH")
        targets = [
            row for row in c142["target_candidate_rows"]
            if row["candidate_id"] == candidate
        ]
        require(
            len(targets) == 1
            and targets[0]["diagnostic_order_status"] == "C036_FULL_ORDER_MATCH_RECORDED",
            "C144_DIAGNOSTIC_ORDER_BINDING_CONFLICT",
        )
        result["diagnostic_order_status"] = targets[0]["diagnostic_order_status"]

        stage = "READ_ORIGINAL_REGRESSION_SPEC"
        original = json.loads(read_bound(spec_path, c078_sha))
        id_fields = ("regression_spec_id", "regression_id", "spec_id")
        rows = [
            (pointer, row)
            for pointer, row in walk(original)
            if any(row.get(key) == spec_id for key in id_fields)
            or pointer.rsplit("/", 1)[-1] == spec_id
        ]
        result["original_spec_candidate_pointers"] = [pointer for pointer, _ in rows]
        if not rows:
            # Keep the exact paths where the ID occurs for a targeted schema correction.
            result["original_spec_id_value_pointers"] = [
                f"{pointer}/{key.replace('~', '~0').replace('/', '~1')}"
                for pointer, row in walk(original)
                for key, value in row.items()
                if isinstance(value, str) and value == spec_id
            ]
        require(len(rows) == 1, f"C144_EXPECTED_ONE_ORIGINAL_SPEC:{len(rows)}")
        pointer, spec = rows[0]
        require(
            not spec.get("candidate_id") or spec["candidate_id"] == candidate,
            "C144_ORIGINAL_SPEC_CANDIDATE_CONFLICT",
        )

        stage = "READ_EXACT_GOLDEN_QUESTION"
        questions = []
        for line_number, line in enumerate(read_bound(golden_path, golden_sha).splitlines(), 1):
            if not line.strip():
                continue
            row = json.loads(line)
            if row["candidate_id"] == candidate:
                questions.append((line_number, row["evaluation_content"]["question"]))
        require(len(questions) == 1, "C144_EXPECTED_ONE_GOLDEN_ROW")
        line_number, question = questions[0]
        require(isinstance(question, str) and bool(question.strip()),
                "C144_EMPTY_GOLDEN_QUESTION")

        result.update(
            status="TARGET_SCENARIO_BOUND_TO_ORIGINAL_SPEC_AND_GOLDEN_QUESTION",
            blocker=None,
            question=question,
            question_sha256=sha(question.encode("utf-8")),
            question_source={
                "path": golden_path.as_posix(),
                "line_number": line_number,
                "json_pointer": "/evaluation_content/question",
            },
            original_spec_source={
                "path": spec_path.as_posix(),
                "json_pointer": pointer,
            },
            original_regression_spec=spec,
            next_product_dependency=(
                "VERIFY_VERSION_MATCHED_PRE_GENERATION_RECEIPT_FOR_THIS_TARGET"
            ),
        )
    except (OSError, ValueError, KeyError, TypeError) as error:
        result.update(
            status="ORIGINAL_TARGET_SCENARIO_BINDING_BLOCKED",
            blocker={"stage": stage, "type": type(error).__name__, "detail": str(error)},
            next_product_dependency="RESOLVE_RECORDED_SOURCE_BINDING_GAP",
        )

    raw = json.dumps(
        result, ensure_ascii=False, sort_keys=True, indent=2, allow_nan=False,
    ).encode("utf-8")
    digest = sha(raw)
    relative = reports / f"C144.R11_original_target_scenario_binding.{digest}.json"
    target = root / relative
    try:
        with target.open("xb") as handle:
            handle.write(raw)
            handle.flush()
            os.fsync(handle.fileno())
    except FileExistsError:
        pass
    require(target.read_bytes() == raw, "C144_REPORT_RELOAD_CONFLICT")
    result["output"] = {
        "path": relative.as_posix(),
        "file_bytes_sha256": digest,
        "reload_validation": "PASS",
    }
    print(json.dumps(result, ensure_ascii=False, indent=2))
    return result


V3_C144 = run_v3_c144()
````

### سل ۱۴۵ — بررسی صلاحیت رسید پیش از تولید برای سناریوی هدف

تعریف اصلی سناریو از C144، تطابق ترتیب تشخیصی از C090 و رسید پیش از تولید
از C141 با هش خوانده می‌شوند. هویت پرونده، هش سؤال و نوع اجرای C141 جداگانه
کنترل می‌شوند.

اگر رسید به سؤال دیگری تعلق داشته باشد، همان اختلاف ثبت می‌شود. در این
سل رتبه یا پاسخ تازه‌ای تولید نمی‌شود و سناریو رأی قبولی یا مردودی نمی‌گیرد.

</div>

````python
# C145 — R11.3 — Check the bound pre-generation receipt for the exact target
def run_v3_c145():
    reports = Path("reports/phase2/quality_improvement/NB22E.lab.v3")
    cwd = Path.cwd()
    root = next((p.resolve() for p in (cwd, *cwd.parents)
                 if (p / reports).is_dir()), None)
    if root is None:
        raise ValueError("C145_PROJECT_ROOT_NOT_FOUND")

    def require(ok, code):
        if not ok:
            raise ValueError(code)

    def sha(raw):
        return hashlib.sha256(raw).hexdigest()

    def load(relative, digest):
        relative = Path(relative)
        path = (root / relative).resolve()
        require(not relative.is_absolute() and path.is_relative_to(root),
                "C145_PATH_OUTSIDE_PROJECT")
        raw = path.read_bytes()
        require(sha(raw) == digest, f"C145_HASH_CONFLICT:{relative}")
        return json.loads(raw)

    pins = {
        "C144": (
            "C144.R11_original_target_scenario_binding",
            "edf453706d80836cb0cfd09b116e36bb7bad3169b1a372e23f0267fd29530570",
        ),
        "C090": (
            "C090.R11_fixed_input_diagnostic_receipts",
            "180520de851dae9fc9265574d36805e60f57e94ae357936db8601969dcd91a47",
        ),
        "C141": (
            "C141.R11_standalone_local_rank_chain",
            "8d45d6d11d8b5fe31863c585b4b5b354265562e47647eb094db415f45e1ca418",
        ),
    }
    documents = {
        cell: load(reports / f"{stem}.{digest}.json", digest)
        for cell, (stem, digest) in pins.items()
    }
    target = documents["C144"]
    diagnostic = documents["C090"]
    run = documents["C141"]
    candidate = target["candidate_id"]
    question_sha = target["question_sha256"]
    spec = target["original_regression_spec"]

    require(
        target["status"] == "TARGET_SCENARIO_BOUND_TO_ORIGINAL_SPEC_AND_GOLDEN_QUESTION"
        and spec["candidate_id"] == candidate
        and spec["regression_spec_id"] == target["regression_spec_id"]
        and spec["kind"] == "SOURCE_BOUND_ANSWER_DECISION_REGRESSION"
        and spec["expected_behavior"] == "ANSWER_SUPPORTED"
        and sha(target["question"].encode("utf-8")) == question_sha,
        "C145_TARGET_SPEC_BINDING_CONFLICT",
    )
    require(
        diagnostic["status"] == "FIXED_INPUT_DIAGNOSTIC_RECEIPTS_RECONSTRUCTED"
        and candidate in diagnostic["c036_order_matched_candidate_ids"]
        and candidate not in diagnostic["rank_origin_unverified_candidate_ids"],
        "C145_DIAGNOSTIC_ORDER_BINDING_CONFLICT",
    )
    require(
        run["status"] == "STANDALONE_LOCAL_RANK_CHAIN_PASSED"
        and run["rank_providers_executed_in_standalone_product_call"] is True
        and run["events"][-3:] == [
            "input_receipt_written",
            "stub_verified_persisted_input",
            "output_receipt_written",
        ],
        "C145_C141_RUN_CONTRACT_CONFLICT",
    )

    ref = run["receipts"]["input"]
    receipt = load(ref["path"], ref["file_bytes_sha256"])
    require(
        receipt["candidate_id"] == run["candidate_id"]
        and receipt["question_sha256"] == run["question_sha256"]
        and receipt["selected_variant"] == "char_rrf"
        and receipt["runtime_identity"]["startup_module_sha256"]
            == run["startup_module_sha256"],
        "C145_C141_INPUT_RECEIPT_CONFLICT",
    )
    same_candidate = receipt["candidate_id"] == candidate
    same_question = receipt["question_sha256"] == question_sha
    require(not (same_candidate and same_question),
            "C145_UNEXPECTED_TARGET_MATCH_REQUIRES_SCOPE_REVIEW")

    result = {
        "cell": "C145",
        "roadmap_action": "R11",
        "roadmap_subtask": "R11.3",
        "status": "TARGET_PRE_GENERATION_RECEIPT_NOT_ESTABLISHED_IN_BOUND_EVIDENCE",
        "input_file_hashes": {
            **{key: digest for key, (_, digest) in pins.items()},
            "C141_input_receipt": ref["file_bytes_sha256"],
        },
        "candidate_id": candidate,
        "question_sha256": question_sha,
        "regression_spec_id": spec["regression_spec_id"],
        "expected_behavior": spec["expected_behavior"],
        "required_limit_disclosed": spec["required_limit_disclosed"],
        "diagnostic_order_status": "C036_FULL_ORDER_MATCH_RECORDED",
        "examined_receipt": {
            "path": ref["path"],
            "candidate_id": receipt["candidate_id"],
            "question_sha256": receipt["question_sha256"],
            "same_candidate": same_candidate,
            "same_question": same_question,
            "generator_kind": "OFFLINE_STUB",
            "disposition": "DEVELOPMENT_PROBE_FOR_DIFFERENT_QUERY",
        },
        "selected_runtime_receipt_verified": False,
        "product_regression_tests_executed": 0,
        "target_case_rankings_reconstructed": 0,
        "new_api_calls": 0,
        "done_item_count": 7,
        "open_item_count": 5,
        "newly_completed_items_from_previous_open_set": [],
        "selected_version_gate": "BLOCKED",
        "next_product_dependency": "BIND_SELECTED_ROUTE_INPUT_ASSETS_FOR_THIS_TARGET_BEFORE_GENERATION",
    }

    raw = json.dumps(result, ensure_ascii=False, sort_keys=True, indent=2,
                     allow_nan=False).encode("utf-8")
    digest = sha(raw)
    relative = reports / f"C145.R11_target_pre_generation_receipt_check.{digest}.json"
    path = root / relative
    try:
        with path.open("xb") as handle:
            handle.write(raw)
            handle.flush()
            os.fsync(handle.fileno())
    except FileExistsError:
        pass
    require(path.read_bytes() == raw, "C145_REPORT_RELOAD_CONFLICT")
    result["output"] = {
        "path": relative.as_posix(),
        "file_bytes_sha256": digest,
        "reload_validation": "PASS",
    }
    print(json.dumps(result, ensure_ascii=False, indent=2))
    return result


V3_C145 = run_v3_c145()
````

### سل ۱۴۶ — اتصال دارایی‌های بازیابی سناریوی پس‌انداز

رسیدهای C144 و C145 هویت سؤال و سناریو را تعیین می‌کنند. گزارش C056
منشأ فایل‌های Golden، Hybrid100 و reranker را مشخص می‌کند. ردیف‌های همان
پرونده با هش فایل و شمارهٔ خط ثبت می‌شوند. عضویت Pool50 با ۵۰ قطعهٔ اول
Hybrid100 و امتیازهای ذخیره‌شده با قرارداد ورودی مقایسه می‌شوند.

تقسیم Development یا Test از ردیف Hybrid خوانده می‌شود تا کش بردار
مربوط در گام بعد احراز شود. این سل هیچ رتبه یا پاسخی تولید نمی‌کند.

</div>

````python
# C146 — R11.3 — Bind historical retrieval inputs for the exact regression target
def run_v3_c146():
    def require(ok, code):
        if not ok:
            raise ValueError(code)

    def sha(raw):
        return hashlib.sha256(raw).hexdigest()

    reports = Path("reports/phase2/quality_improvement/NB22E.lab.v3")
    cwd = Path.cwd()
    root = next(
        (p.resolve() for p in (cwd, *cwd.parents) if (p / reports).is_dir()),
        None,
    )
    require(root is not None, "C146_PROJECT_ROOT_NOT_FOUND")

    def read_bound(relative, digest):
        relative = Path(relative)
        path = (root / relative).resolve()
        require(
            not relative.is_absolute() and path.is_relative_to(root),
            "C146_PATH_OUTSIDE_PROJECT",
        )
        raw = path.read_bytes()
        require(sha(raw) == digest, f"C146_HASH_CONFLICT:{relative}")
        return raw

    def load_report(stem, digest):
        return json.loads(read_bound(reports / f"{stem}.{digest}.json", digest))

    c144_sha = "edf453706d80836cb0cfd09b116e36bb7bad3169b1a372e23f0267fd29530570"
    c145_sha = "2dceef6a676b65d57dafad74231acaede2c85459729e6bb243b465b14e96a41c"
    c056_sha = "ff6d6e136043929e0156e43f6ea75e65bd64f40ab7bcdafef90a4ba40901de22"

    result = {
        "cell": "C146",
        "roadmap_action": "R11",
        "roadmap_subtask": "R11.3",
        "scope": "ONE_TARGET_HISTORICAL_RETRIEVAL_INPUT_BINDING",
        "input_file_hashes": {
            "C144": c144_sha, "C145": c145_sha, "NB22_C056": c056_sha,
        },
        "historical_retrieval_inputs_bound": False,
        "query_vector_bound": False,
        "selected_runtime_receipt_verified": False,
        "product_regression_tests_executed": 0,
        "target_case_rankings_reconstructed": 0,
        "rank_reconstruction_for_unverified_candidates": "DEFERRED",
        "qdrant_queries_attempted": 0,
        "new_local_pairs_scored": 0,
        "new_api_calls": 0,
        "done_item_count": 7,
        "open_item_count": 5,
        "newly_completed_items_from_previous_open_set": [],
        "historical_gate": "FAIL",
        "selected_version_gate": "BLOCKED",
    }

    stage = "BIND_TARGET_SCENARIO"
    try:
        target = load_report("C144.R11_original_target_scenario_binding", c144_sha)
        prior = load_report("C145.R11_target_pre_generation_receipt_check", c145_sha)
        candidate = target["candidate_id"]
        question = target["question"]
        question_sha = target["question_sha256"]
        spec_id = target["regression_spec_id"]
        require(
            target["status"] == "TARGET_SCENARIO_BOUND_TO_ORIGINAL_SPEC_AND_GOLDEN_QUESTION"
            and prior["status"] == "TARGET_PRE_GENERATION_RECEIPT_NOT_ESTABLISHED_IN_BOUND_EVIDENCE"
            and prior["input_file_hashes"]["C144"] == c144_sha
            and (prior["candidate_id"], prior["question_sha256"], prior["regression_spec_id"])
                == (candidate, question_sha, spec_id)
            and sha(question.encode("utf-8")) == question_sha
            and target["original_regression_spec"]["expected_behavior"]
                == prior["expected_behavior"] == "ANSWER_SUPPORTED",
            "C146_TARGET_BINDING_CONFLICT",
        )
        result.update(
            candidate_id=candidate,
            question_sha256=question_sha,
            regression_spec_id=spec_id,
            expected_behavior=prior["expected_behavior"],
        )

        stage = "READ_PINNED_C056_INPUTS"
        c056 = json.loads(read_bound(
            Path("reports/phase2/release/NB22.C056.retrieval_verification")
            / f"C056.independent_retrieval_metric_verification.{c056_sha}.json",
            c056_sha,
        ))
        require(
            c056["independent_recomputation_complete"] is True
            and c056["all_recorded_metric_comparisons_match"] is True,
            "C146_VERIFIED_C056_REQUIRED",
        )
        bindings = c056["input_file_receipts"]
        require(
            bindings["golden"]["file_bytes_sha256"] == target["input_file_hashes"]["golden"],
            "C146_GOLDEN_VERSION_CONFLICT",
        )

        row_receipts = {}

        def target_row(role):
            binding = bindings[role]
            path = binding["path"]
            digest = binding["file_bytes_sha256"]
            matches = []
            for number, line in enumerate(read_bound(path, digest).splitlines(), 1):
                if line.strip():
                    row = json.loads(line)
                    if row["candidate_id"] == candidate:
                        matches.append((number, line, row))
            require(len(matches) == 1, f"C146_EXPECTED_ONE_TARGET_ROW:{role}:{len(matches)}")
            number, line, row = matches[0]
            row_receipts[role] = {
                "path": path,
                "file_bytes_sha256": digest,
                "line_number": number,
                "row_sha256_excluding_line_ending": sha(line),
            }
            return row

        golden = target_row("golden")
        require(
            golden["evaluation_content"]["question"] == question
            and row_receipts["golden"]["line_number"] == target["question_source"]["line_number"],
            "C146_GOLDEN_QUESTION_BINDING_CONFLICT",
        )

        stage = "VERIFY_TARGET_HYBRID100_AND_POOL50"
        hybrid_row = target_row("hybrid")
        pool_row = target_row("reranker")
        split = hybrid_row["split"]
        require(split in {"development", "test"}, "C146_UNSUPPORTED_SPLIT")
        require(pool_row.get("split", split) == split, "C146_SPLIT_CONFLICT")
        result["target_split"] = split

        def ordered_ids(row, depth):
            candidates = row["candidates"]
            require(
                [entry["rank"] for entry in candidates] == list(range(1, depth + 1)),
                f"C146_RANK_ORDER_CONFLICT:{depth}",
            )
            ids = [entry["chunk_id"] for entry in candidates]
            require(
                all(isinstance(cid, str) and cid for cid in ids)
                and len(set(ids)) == depth,
                f"C146_CHUNK_ID_CONFLICT:{depth}",
            )
            return ids

        hybrid = ordered_ids(hybrid_row, 100)
        pool = ordered_ids(pool_row, 50)
        require(set(pool) == set(hybrid[:50]), "C146_POOL50_MEMBERSHIP_CONFLICT")
        scores = {
            row["chunk_id"]: float(row["reranker_score"])
            for row in pool_row["candidates"]
        }
        require(all(math.isfinite(score) for score in scores.values()),
                "C146_NONFINITE_POOL50_SCORE")

        result.update(
            status="TARGET_HISTORICAL_HYBRID100_AND_POOL50_BOUND",
            blocker=None,
            historical_retrieval_inputs_bound=True,
            source_rows=row_receipts,
            hybrid100_order=hybrid,
            pool50_order=pool,
            pool50_scores_by_chunk_id=scores,
            hybrid_depth=len(hybrid),
            pool50_score_count=len(scores),
            pool50_membership_matches_hybrid_first50=True,
            pool50_order_equals_hybrid_first50=(pool == hybrid[:50]),
            next_product_dependency=f"VERIFY_PINNED_{split.upper()}_QUERY_VECTOR_FOR_TARGET",
        )
    except (OSError, ValueError, KeyError, TypeError) as error:
        result.update(
            status="TARGET_RETRIEVAL_INPUT_BINDING_BLOCKED",
            blocker={"stage": stage, "type": type(error).__name__, "detail": str(error)},
            next_product_dependency="RESOLVE_RECORDED_TARGET_INPUT_GAP",
        )

    raw = json.dumps(
        result, ensure_ascii=False, sort_keys=True, indent=2, allow_nan=False,
    ).encode("utf-8")
    digest = sha(raw)
    relative = reports / f"C146.R11_target_retrieval_input_binding.{digest}.json"
    path = root / relative
    try:
        with path.open("xb") as handle:
            handle.write(raw)
            handle.flush()
            os.fsync(handle.fileno())
    except FileExistsError:
        pass
    require(path.read_bytes() == raw, "C146_REPORT_RELOAD_CONFLICT")
    result["output"] = {
        "path": relative.as_posix(),
        "file_bytes_sha256": digest,
        "reload_validation": "PASS",
    }
    details = {"hybrid100_order", "pool50_order", "pool50_scores_by_chunk_id"}
    print(json.dumps(
        {key: value for key, value in result.items() if key not in details},
        ensure_ascii=False, indent=2,
    ))
    return result


V3_C146 = run_v3_c146()
````

### سل ۱۴۷ — اتصال بردار کش‌شدهٔ Test به سؤال هدف

ترتیب شناسه‌های Test، هش مجموعهٔ سؤال‌ها و کلید کش مطابق قرارداد NB16
بازسازی و با متادیتای کش مقایسه می‌شوند. بردار همان سؤال از ردیف متناظر
خوانده می‌شود و شکل، نوع داده، مقادیر متناهی و نُرم آن کنترل می‌شوند.

کش فقط خوانده می‌شود. مدل embedding، Qdrant و مولد پاسخ اجرا نمی‌شوند.

</div>

````python
# C147 — R11.3 — Bind the cached Test vector for the exact target
def run_v3_c147():
    def require(ok, code):
        if not ok:
            raise ValueError(code)

    def sha(raw):
        return hashlib.sha256(raw).hexdigest()

    def canonical_sha(value):
        return sha(json.dumps(
            value, ensure_ascii=False, sort_keys=True, separators=(",", ":"),
        ).encode("utf-8"))

    reports = Path("reports/phase2/quality_improvement/NB22E.lab.v3")
    cwd = Path.cwd()
    root = next((p.resolve() for p in (cwd, *cwd.parents)
                 if (p / reports).is_dir()), None)
    require(root is not None, "C147_PROJECT_ROOT_NOT_FOUND")

    def project_path(relative):
        relative = Path(relative)
        path = (root / relative).resolve()
        require(not relative.is_absolute() and path.is_relative_to(root),
                "C147_PATH_OUTSIDE_PROJECT")
        return path

    def read_bound(relative, digest):
        raw = project_path(relative).read_bytes()
        require(sha(raw) == digest, f"C147_HASH_CONFLICT:{relative}")
        return raw

    c146_sha = "226aca0c92a8bd7f23f06b6714c7d427f65b46799a2b7fe6e9456cc7710eeb2e"
    nb16_sha = "40699ca54c57fcef40472df6b6fc42a54f74dc2e9bdc9bb40e0ccc93bb7ad6a7"
    cache = Path("artifacts/phase2/hybrid_retrieval/query_embedding_cache_v1")
    metadata_path = cache / "test_query_embeddings.json"
    vector_path = cache / "test_query_embeddings.npy"
    result = {
        "cell": "C147",
        "roadmap_action": "R11",
        "roadmap_subtask": "R11.3",
        "input_file_hashes": {"C146": c146_sha, "NB16": nb16_sha},
        "scope": "ONE_TARGET_EXISTING_TEST_QUERY_VECTOR",
        "query_vector_bound": False,
        "selected_runtime_receipt_verified": False,
        "embedding_calls": 0,
        "qdrant_queries_attempted": 0,
        "new_api_calls": 0,
        "target_case_rankings_reconstructed": 0,
        "product_regression_tests_executed": 0,
        "done_item_count": 7,
        "open_item_count": 5,
        "selected_version_gate": "BLOCKED",
    }

    stage = "BIND_TARGET_AND_SOURCE"
    try:
        prior = json.loads(read_bound(
            reports / f"C146.R11_target_retrieval_input_binding.{c146_sha}.json",
            c146_sha,
        ))
        require(
            prior["status"] == "TARGET_HISTORICAL_HYBRID100_AND_POOL50_BOUND"
            and prior["historical_retrieval_inputs_bound"] is True
            and prior["target_split"] == "test",
            "C147_BOUND_TEST_TARGET_REQUIRED",
        )
        read_bound("notebooks/phase2/16_hybrid_retrieval.ipynb", nb16_sha)
        candidate = prior["candidate_id"]
        result.update(
            candidate_id=candidate,
            question_sha256=prior["question_sha256"],
            regression_spec_id=prior["regression_spec_id"],
            target_split="test",
        )

        stage = "REBUILD_PINNED_TEST_CACHE_CONTRACT"
        hybrid_ref = prior["source_rows"]["hybrid"]
        golden_ref = prior["source_rows"]["golden"]
        hybrid_rows = [
            json.loads(line)
            for line in read_bound(
                hybrid_ref["path"], hybrid_ref["file_bytes_sha256"]
            ).splitlines() if line.strip()
        ]
        ids = sorted(row["candidate_id"] for row in hybrid_rows if row["split"] == "test")
        require(len(ids) == len(set(ids)) == 1149 and candidate in ids,
                "C147_TEST_POPULATION_CONFLICT")
        wanted = set(ids)
        questions = {}
        for line in read_bound(
            golden_ref["path"], golden_ref["file_bytes_sha256"],
        ).splitlines():
            if not line.strip():
                continue
            row = json.loads(line)
            cid = row["candidate_id"]
            if cid in wanted:
                text = row["evaluation_content"]["question"]
                require(cid not in questions and isinstance(text, str) and text.strip(),
                        f"C147_INVALID_GOLDEN_QUESTION:{cid}")
                if cid == candidate:
                    require(sha(text.encode("utf-8")) == prior["question_sha256"],
                            "C147_EXACT_TARGET_QUESTION_HASH_CONFLICT")
                questions[cid] = text.strip()

        require(set(questions) == wanted, "C147_TEST_QUESTION_POPULATION_CONFLICT")
        query_set_sha = canonical_sha([
            {"candidate_id": cid, "question": questions[cid]} for cid in ids
        ])
        contract = {
            "split": "test",
            "query_set_sha256": query_set_sha,
            "model_id": "intfloat/multilingual-e5-small",
            "model_revision": "614241f622f53c4eeff9890bdc4f31cfecc418b3",
            "query_prefix": "query:",
            "normalize_embeddings": True,
        }
        cache_key = canonical_sha(contract)

        stage = "VERIFY_EXISTING_TEST_CACHE"
        metadata_raw = project_path(metadata_path).read_bytes()
        metadata = json.loads(metadata_raw)
        require(
            metadata["artifact_type"] == "NB16_TEST_QUERY_EMBEDDING_CACHE"
            and metadata["schema_version"] == "1.0"
            and metadata["candidate_ids"] == ids
            and metadata["cache_key"] == cache_key
            and metadata["query_count"] == 1149
            and metadata["shape"] == [1149, 384]
            and metadata["dtype"] == "float32"
            and all(metadata.get(key) == value for key, value in contract.items()),
            "C147_TEST_CACHE_METADATA_CONFLICT",
        )

        array_path = project_path(vector_path)
        vector_file_sha = sha(array_path.read_bytes())
        array = np.load(array_path, allow_pickle=False)
        require(
            sha(array_path.read_bytes()) == vector_file_sha
            and project_path(metadata_path).read_bytes() == metadata_raw,
            "C147_CACHE_CHANGED_DURING_READ",
        )
        require(array.shape == (1149, 384) and array.dtype == np.float32,
                "C147_TEST_CACHE_ARRAY_CONFLICT")
        row_index = ids.index(candidate)
        vector = array[row_index]
        norm = float(np.linalg.norm(vector))
        require(bool(np.isfinite(vector).all()) and bool(np.isclose(norm, 1.0)),
                "C147_INVALID_TARGET_VECTOR")

        result.update(
            status="TARGET_TEST_QUERY_VECTOR_BOUND",
            blocker=None,
            query_vector_bound=True,
            cache_row_index_zero_based=row_index,
            cache_key=cache_key,
            test_query_set_sha256=query_set_sha,
            embedded_question_sha256=sha(questions[candidate].encode("utf-8")),
            query_vector_sha256=sha(vector.tobytes()),
            query_vector_norm=norm,
            cache_files={
                "metadata": {
                    "path": metadata_path.as_posix(),
                    "file_bytes_sha256": sha(metadata_raw),
                },
                "array": {
                    "path": vector_path.as_posix(),
                    "file_bytes_sha256": vector_file_sha,
                },
            },
            next_product_dependency="RUN_TARGET_HYBRID100_WITH_BOUND_TEST_VECTOR_AND_COMPARE",
        )
    except (OSError, ValueError, KeyError, TypeError, NameError) as error:
        result.update(
            status="TARGET_TEST_QUERY_VECTOR_BINDING_BLOCKED",
            blocker={"stage": stage, "type": type(error).__name__, "detail": str(error)},
            next_product_dependency="RESOLVE_RECORDED_TEST_CACHE_GAP",
        )

    raw = json.dumps(result, ensure_ascii=False, sort_keys=True, indent=2,
                     allow_nan=False).encode("utf-8")
    digest = sha(raw)
    relative = reports / f"C147.R11_target_test_query_vector_binding.{digest}.json"
    path = project_path(relative)
    try:
        with path.open("xb") as handle:
            handle.write(raw)
            handle.flush()
            os.fsync(handle.fileno())
    except FileExistsError:
        pass
    require(path.read_bytes() == raw, "C147_REPORT_RELOAD_CONFLICT")

    result["output"] = {
        "path": relative.as_posix(),
        "file_bytes_sha256": digest,
        "reload_validation": "PASS",
    }
    print(json.dumps(result, ensure_ascii=False, indent=2))
    return result


V3_C147 = run_v3_c147()
````

````python
# C148 — R11.3 — Run one local Hybrid100 query using the bound Test vector
def run_v3_c148():
    def require(ok, code):
        if not ok:
            raise ValueError(code)

    def sha(raw):
        return hashlib.sha256(raw).hexdigest()

    reports = Path("reports/phase2/quality_improvement/NB22E.lab.v3")
    cwd = Path.cwd()
    root = next(
        (p.resolve() for p in (cwd, *cwd.parents) if (p / reports).is_dir()),
        None,
    )
    require(root is not None, "C148_PROJECT_ROOT_NOT_FOUND")

    def read_bound(relative, digest):
        relative = Path(relative)
        path = (root / relative).resolve()
        require(not relative.is_absolute() and path.is_relative_to(root),
                "C148_PATH_OUTSIDE_PROJECT")
        raw = path.read_bytes()
        require(sha(raw) == digest, f"C148_HASH_CONFLICT:{relative}")
        return raw

    c146_sha = "226aca0c92a8bd7f23f06b6714c7d427f65b46799a2b7fe6e9456cc7710eeb2e"
    c147_sha = "4027a22f2cc0c1f38393784174267ba6867f1596aad5ef37dad034c7c0283321"
    target = json.loads(read_bound(
        reports / f"C146.R11_target_retrieval_input_binding.{c146_sha}.json",
        c146_sha,
    ))
    cache = json.loads(read_bound(
        reports / f"C147.R11_target_test_query_vector_binding.{c147_sha}.json",
        c147_sha,
    ))
    identity_fields = ("candidate_id", "question_sha256", "regression_spec_id")
    require(
        target["status"] == "TARGET_HISTORICAL_HYBRID100_AND_POOL50_BOUND"
        and cache["status"] == "TARGET_TEST_QUERY_VECTOR_BOUND"
        and cache["query_vector_bound"] is True
        and cache["input_file_hashes"]["C146"] == c146_sha
        and target["target_split"] == cache["target_split"] == "test"
        and all(target[key] == cache[key] for key in identity_fields),
        "C148_TARGET_AND_CACHE_BINDING_CONFLICT",
    )

    def load_target_vector(scope, result):
        candidate = target["candidate_id"]
        require(candidate in scope["test_candidate_ids"], "C148_NB16_TEST_ID_MISSING")
        question = scope["golden_record_by_candidate_id"][candidate][
            "evaluation_content"
        ]["question"]
        require(sha(question.encode("utf-8")) == target["question_sha256"],
                "C148_GOLDEN_QUESTION_CONFLICT")
        question = question.strip()
        require(sha(question.encode("utf-8")) == cache["embedded_question_sha256"],
                "C148_EMBEDDED_QUESTION_CONFLICT")

        refs = cache["cache_files"]
        meta_ref, array_ref = refs["metadata"], refs["array"]
        metadata = json.loads(read_bound(meta_ref["path"], meta_ref["file_bytes_sha256"]))
        index = cache["cache_row_index_zero_based"]
        ids = metadata["candidate_ids"]
        require(
            metadata["cache_key"] == cache["cache_key"]
            and metadata["query_set_sha256"] == cache["test_query_set_sha256"]
            and ids == sorted(str(cid) for cid in scope["test_candidate_ids"])
            and type(index) is int and 0 <= index < len(ids)
            and ids[index] == candidate,
            "C148_TEST_CACHE_ROW_CONFLICT",
        )

        array_path = (root / array_ref["path"]).resolve()
        read_bound(array_ref["path"], array_ref["file_bytes_sha256"])
        np_local = scope["np"]
        array = np_local.load(array_path, allow_pickle=False)
        read_bound(array_ref["path"], array_ref["file_bytes_sha256"])
        require(array.shape == (1149, 384) and array.dtype == np_local.float32,
                "C148_TEST_CACHE_ARRAY_CONFLICT")
        vector = array[index].copy()
        require(
            bool(np_local.isfinite(vector).all())
            and bool(np_local.isclose(np_local.linalg.norm(vector), 1.0))
            and sha(vector.tobytes()) == cache["query_vector_sha256"],
            "C148_TEST_VECTOR_CONFLICT",
        )
        result["input_file_hashes"].update(
            test_cache_metadata=meta_ref["file_bytes_sha256"],
            test_cache_array=array_ref["file_bytes_sha256"],
        )
        result.update(
            query_vector_bound=True,
            cache_row_index_zero_based=index,
            query_vector_sha256=cache["query_vector_sha256"],
        )
        return question, vector

    def record_outcome(result):
        result.update(
            regression_spec_id=target["regression_spec_id"],
            target_split="test",
            pool50_membership_matches_computed_first50=None,
        )
        observed = result.get("observed_hybrid100_order")
        if observed is not None:
            result["pool50_membership_matches_computed_first50"] = (
                len(observed) >= 50
                and set(observed[:50]) == set(target["pool50_order"])
            )

        if result["full_hybrid100_order_matches"]:
            result["status"] = "TARGET_TEST_HYBRID100_ORDER_MATCHED"
            result["next_product_dependency"] = (
                "BIND_TARGET_LOCAL100_AND_CHAR_RRF_BEFORE_GENERATION"
            )
        elif observed is not None:
            result["status"] = "TARGET_TEST_HYBRID100_ORDER_DIFFERS"
            result["next_product_dependency"] = "RESOLVE_TARGET_HYBRID100_ORDER_DIFFERENCE"
        else:
            result["status"] = "TARGET_TEST_HYBRID100_PROBE_BLOCKED"
            result["next_product_dependency"] = "RESOLVE_RECORDED_TARGET_HYBRID_BLOCKER"

    # Reuse the successful C126 setup and scorer. Replace only the query binding.
    history = globals().get("In")
    require(isinstance(history, list), "C148_IPYTHON_HISTORY_REQUIRED")
    entry = next(
        (item for item in reversed(history)
         if isinstance(item, str) and "def run_v3_c126():" in item.splitlines()),
        None,
    )
    require(entry is not None, "C148_C126_SOURCE_REQUIRED")
    definitions = [
        node for node in ast.parse(entry).body
        if isinstance(node, ast.FunctionDef) and node.name == "run_v3_c126"
    ]
    require(len(definitions) == 1, "C148_C126_SOURCE_AMBIGUOUS")
    source = ast.get_source_segment(entry, definitions[0])

    def replace_once(old, new):
        nonlocal source
        require(source.count(old) == 1, f"C148_C126_SOURCE_CONTRACT_CHANGED:{old}")
        source = source.replace(old, new, 1)

    replace_once(
        '''        prior = json.loads(read_bound(
            reports / f"C120.R11_development_local100_input.{c120_sha}.json",
            c120_sha,
        ))''',
        "        prior = c148_target",
    )
    replace_once(
        '"ONE_DEVELOPMENT_LOCAL100_INPUT_BUILT"',
        '"TARGET_HISTORICAL_HYBRID100_AND_POOL50_BOUND"',
    )
    replace_once(
        '"input_file_hashes": {"NB16": nb_sha, "C120": c120_sha},',
        '"input_file_hashes": {"NB16": nb_sha, "C146": c148_c146_sha, '
        '"C147": c148_c147_sha},\n        "query_vector_bound": False,',
    )
    begin = '        stage = "LOAD_EXISTING_DEVELOPMENT_CACHE"'
    end = '        stage = "OPEN_AND_VERIFY_LOCAL_QDRANT"'
    require(source.count(begin) == source.count(end) == 1,
            "C148_C126_CACHE_BOUNDARY_CONFLICT")
    start_index = source.index(begin)
    end_index = source.index(end)
    require(start_index < end_index, "C148_C126_CACHE_BOUNDARY_ORDER_CONFLICT")
    source = (
        source[:start_index]
        + '        stage = "LOAD_BOUND_TEST_VECTOR"\n'
          '        question, vector = c148_load_target_vector(scope, result)\n\n'
        + source[end_index:]
    )

    replace_once(
        '    raw = json.dumps(\n        result,',
        '    c148_record_outcome(result)\n\n    raw = json.dumps(\n        result,',
    )
    replace_once('"cell": "C126"', '"cell": "C148"')
    replace_once(
        '"ONE_DEVELOPMENT_QUERY_LOCAL_NB16_ASSETS"',
        '"ONE_REGRESSION_TARGET_TEST_QUERY_LOCAL_NB16_ASSETS"',
    )
    replace_once("C126.R11_local_hybrid100_probe", "C148.R11_target_test_hybrid100_probe")
    replace_once("def run_v3_c126():", "def run_v3_c148_probe():")
    source = source.replace("C126_", "C148_")

    scope = dict(globals())
    scope.update(
        c148_target=target,
        c148_c146_sha=c146_sha,
        c148_c147_sha=c147_sha,
        c148_load_target_vector=load_target_vector,
        c148_record_outcome=record_outcome,
    )
    exec(compile(source, "<C148:adapted-C126>", "exec"), scope)
    return scope["run_v3_c148_probe"]()


V3_C148 = run_v3_c148()
````

### سل ۱۴۹ — ساخت ترتیب char_rrf برای سناریوی هدف Test

ترتیب Hybrid100 از C148 و امتیازهای پایه از C146 خوانده می‌شوند.
۵۰ جفت باقی‌مانده با مدل محلی پین‌شده امتیاز می‌گیرند. سپس Local100،

این سل ورودی اجرای بعدی را آماده می‌کند. پاسخ تولید نمی‌شود و اجرای
رگرسیون یا صدور رسید محصول در این مرحله ثبت نمی‌شود.

````python
# C149 — R11.3 — Compute one target's Local100, Quota100 and char_rrf
def run_v3_c149():
    def require(ok, code):
        if not ok:
            raise ValueError(code)

    def sha(raw):
        return hashlib.sha256(raw).hexdigest()

    def text_sha(value):
        return sha(value.encode("utf-8"))

    reports = Path("reports/phase2/quality_improvement/NB22E.lab.v3")
    cwd = Path.cwd()
    root = next((p.resolve() for p in (cwd, *cwd.parents)
                 if (p / reports).is_dir()), None)
    require(root is not None, "C149_PROJECT_ROOT_NOT_FOUND")

    def read_bound(relative, digest):
        relative = Path(relative)
        path = (root / relative).resolve()
        require(not relative.is_absolute() and path.is_relative_to(root),
                "C149_PATH_OUTSIDE_PROJECT")
        raw = path.read_bytes()
        require(sha(raw) == digest, f"C149_HASH_CONFLICT:{relative}")
        return raw

    pins = {
        "C144": (
            "C144.R11_original_target_scenario_binding",
            "edf453706d80836cb0cfd09b116e36bb7bad3169b1a372e23f0267fd29530570",
        ),
        "C146": (
            "C146.R11_target_retrieval_input_binding",
            "226aca0c92a8bd7f23f06b6714c7d427f65b46799a2b7fe6e9456cc7710eeb2e",
        ),
        "C148": (
            "C148.R11_target_test_hybrid100_probe",
            "8352cfa49bedc99c50335f57a66051a3613bdb1cf3618200dff771a8c1d654f4",
        ),
        "C118": (
            "C118.R11_local100_model_preflight",
            "21aee5d3199de8d81afc83b58f96ff67df21fc47b019132c52b8d67235524c4d",
        ),
        "C119": (
            "C119.R11_local100_four_pair_parity",
            "967178107d6a4fab03ecb5ffd74f79ceebfd716dc265d9726339164d06a3647b",
        ),
    }
    result = {
        "cell": "C149",
        "roadmap_action": "R11",
        "roadmap_subtask": "R11.3",
        "scope": "ONE_TEST_TARGET_PRE_GENERATION_RANK_INPUTS",
        "input_file_hashes": {key: digest for key, (_, digest) in pins.items()},
        "qdrant_queries_attempted": 0,
        "new_local_pairs_scored": 0,
        "model_inference_calls_attempted": 0,
        "new_api_calls": 0,
        "product_regression_tests_executed": 0,
        "selected_runtime_receipt_verified": False,
        "rank_reconstruction_for_unverified_candidates": "DEFERRED",
        "done_item_count": 7,
        "open_item_count": 5,
        "selected_version_gate": "BLOCKED",
    }

    stage = "BIND_TARGET_INPUTS"
    try:
        docs = {
            key: json.loads(read_bound(reports / f"{stem}.{digest}.json", digest))
            for key, (stem, digest) in pins.items()
        }
        spec, target, probe = (docs[key] for key in ("C144", "C146", "C148"))
        identity = ("candidate_id", "question_sha256", "regression_spec_id")
        require(
            spec["status"] == "TARGET_SCENARIO_BOUND_TO_ORIGINAL_SPEC_AND_GOLDEN_QUESTION"
            and target["historical_retrieval_inputs_bound"] is True
            and probe["status"] == "TARGET_TEST_HYBRID100_ORDER_MATCHED"
            and probe["query_vector_bound"] is True
            and probe["input_file_hashes"]["C146"] == pins["C146"][1]
            and target["input_file_hashes"]["C144"] == pins["C144"][1]
            and probe["target_split"] == target["target_split"] == "test"
            and all(spec[key] == target[key] == probe[key] for key in identity),
            "C149_TARGET_RECEIPT_CONFLICT",
        )
        result.update({key: target[key] for key in identity})
        question = spec["question"]
        require(text_sha(question) == target["question_sha256"],
                "C149_QUESTION_HASH_CONFLICT")
        # NB22B and NB22D both consume stripped Golden questions.
        question = question.strip()
        result["ranker_question_sha256"] = text_sha(question)

        hybrid = probe["observed_hybrid100_order"]
        pool = target["pool50_order"]
        scores = dict(target["pool50_scores_by_chunk_id"])
        require(
            hybrid == target["hybrid100_order"]
            and len(hybrid) == len(set(hybrid)) == 100
            and len(pool) == len(set(pool)) == 50
            and set(pool) == set(hybrid[:50]) == set(scores)
            and all(math.isfinite(score) for score in scores.values()),
            "C149_HYBRID100_OR_POOL50_CONFLICT",
        )

        stage = "LOAD_PINNED_CORPUS"
        corpus_path = "data/phase2/chunks/nb11_benchmark_v1/baseline_phase1_char_safe_v1.jsonl"
        corpus_sha = "3891c4ecad9332ba753cd605bbccd99da9d063f64934a30ee39e495349cffb50"
        chunks = {}
        for line in read_bound(corpus_path, corpus_sha).splitlines():
            if line.strip():
                row = json.loads(line)
                cid = row["chunk_id"]
                require(cid not in chunks, "C149_DUPLICATE_CHUNK")
                chunks[cid] = row
        require(len(chunks) == 3488 and set(hybrid) <= chunks.keys(),
                "C149_CORPUS_OR_HYBRID_MEMBERSHIP_CONFLICT")
        result["input_file_hashes"]["chunks"] = corpus_sha

        stage = "LOAD_PINNED_LOCAL_MODEL"
        parity = docs["C119"]
        config = docs["C118"]["expected_runtime"]
        require(
            parity["score_parity_verified"] is True
            and parity["input_file_hashes"]["C118"] == pins["C118"][1]
            and (config["batch_size"], config["maximum_sequence_length"]) == (4, 512),
            "C149_RUNTIME_PARITY_OR_CONFIG_CONFLICT",
        )
        runtime = globals().get("ort") or globals().get("onnxruntime")
        require(runtime is not None and runtime.__version__ == config["onnxruntime_version"],
                "C149_ONNXRUNTIME_VERSION_CONFLICT")
        cache_root = Path(
            os.environ.get("HF_HUB_CACHE")
            or str(Path(os.environ.get("HF_HOME", str(Path.home() / ".cache/huggingface"))) / "hub")
        )
        snapshot = (
            cache_root / ("models--" + config["model_id"].replace("/", "--"))
            / "snapshots" / config["model_revision"]
        )
        model = snapshot / "onnx/model_quint8_avx2.onnx"
        require(sha(model.read_bytes()) == config["onnx_sha256"],
                "C149_ONNX_MODEL_HASH_CONFLICT")
        result["input_file_hashes"]["onnx_model"] = config["onnx_sha256"]
        tokenizer = AutoTokenizer.from_pretrained(
            str(snapshot), local_files_only=True, trust_remote_code=False,
        )
        options = runtime.SessionOptions()
        options.intra_op_num_threads = 4
        options.inter_op_num_threads = 1
        options.execution_mode = runtime.ExecutionMode.ORT_SEQUENTIAL
        options.graph_optimization_level = runtime.GraphOptimizationLevel.ORT_ENABLE_ALL
        session = runtime.InferenceSession(
            str(model), sess_options=options, providers=["CPUExecutionProvider"],
        )
        names = {item.name for item in session.get_inputs()}
        require({"input_ids", "attention_mask"} <= names, "C149_ONNX_INPUT_CONFLICT")

        stage = "SCORE_TARGET_MISSING_50"
        missing = [cid for cid in hybrid if cid not in scores]
        require(len(missing) == 50, "C149_MISSING_PAIR_COUNT_CONFLICT")
        for start in range(0, len(missing), 4):
            batch = missing[start:start + 4]
            encoded = tokenizer(
                [question] * len(batch), [chunks[cid]["content"] for cid in batch],
                padding=True, truncation=True, max_length=512, return_tensors="np",
            )
            require(names <= set(encoded), "C149_TOKENIZER_ONNX_INPUT_MISMATCH")
            feed = {name: np.asarray(encoded[name], dtype=np.int64) for name in names}
            result["model_inference_calls_attempted"] += 1
            values = np.asarray(session.run(None, feed)[0], dtype=np.float64).reshape(-1)
            require(values.shape == (len(batch),) and bool(np.isfinite(values).all()),
                    "C149_INVALID_LOCAL_SCORES")
            scores.update({cid: float(value) for cid, value in zip(batch, values)})
            result["new_local_pairs_scored"] += len(batch)

        local = sorted(hybrid, key=lambda cid: (-scores[cid], cid))

        stage = "RUN_PINNED_QUOTA100_AND_CHAR_RRF"

        def load_functions(path, digest, index, names, config):
            notebook = json.loads(read_bound(path, digest))
            source = "".join(notebook["cells"][index]["source"])
            nodes = [
                node for node in ast.parse(source).body
                if isinstance(node, ast.FunctionDef) and node.name in names
            ]
            require(len(nodes) == len(names) and {node.name for node in nodes} == names,
                    f"C149_PINNED_FUNCTIONS_MISSING:{path}")
            scope = dict(globals())
            scope.update(require=require, CONFIG=config)
            exec(compile(ast.Module(body=nodes, type_ignores=[]), path, "exec"), scope)
            return scope

        c_path = "notebooks/phase2/22c_retrieval_diagnosis_and_improvement.ipynb"
        c_sha = "e658b973c057d945add9a9f7c96c6a98b36aae1b2fca70cf958925df4de7315c"
        c_scope = load_functions(c_path, c_sha, 6, {"derived_order"}, {})
        quota = c_scope["derived_order"](
            hybrid, local, pool,
            {"family": "quota100", "k": 10, "weight": 0.75, "newcomer_limit": 1},
        )

        d_path = "notebooks/phase2/22d_full_corpus_retrieval_expansion.ipynb"
        d_sha = "fff5bd072a4977820882045451b61d66f8621ef9a4918775e4534800f0ae1480"
        d_scope = load_functions(
            d_path, d_sha, 12,
            {"normalized_words", "char_tokens", "build_index", "lexical_search",
             "lexical_fusion"},
            {"char_max_df_fraction": 0.30, "bm25": {"k1": 1.2, "b": 0.3},
             "rrf_k": 10, "incumbent_rank_weight": 0.75},
        )
        char_index = d_scope["build_index"](chunks, "char")
        channel = d_scope["lexical_search"](char_index, question, "char")
        fused = d_scope["lexical_fusion"](
            quota, {"char": channel}, {"channels": ["char"]},
        )
        top10 = fused[:10]
        require(
            len(quota) == len(set(quota)) == 100 and set(quota) == set(hybrid)
            and len(top10) == len(set(top10)) == 10
            and set(fused) == set(quota) | set(channel),
            "C149_FINAL_ORDER_CONTRACT_CONFLICT",
        )
        result["input_file_hashes"].update(NB22C=c_sha, NB22D=d_sha)
        result.update(
            status="TARGET_TEST_CHAR_RRF_RANK_INPUTS_BUILT",
            blocker=None,
            hybrid100_order=hybrid,
            pool50_order=pool,
            local100_order=local,
            local100_scores_by_chunk_id=scores,
            quota100_order=quota,
            char_channel_order=channel,
            char_rrf_order=fused,
            char_rrf_top10=top10,
            char_rrf_top10_text_sha256=[text_sha(chunks[cid]["content"]) for cid in top10],
            char_channel_depth=len(channel),
            char_rrf_depth=len(fused),
            reused_pool50_scores=50,
            next_product_dependency="EXERCISE_OPT_IN_PRODUCT_ROUTE_AND_CAPTURE_TARGET_INPUT_RECEIPT",
        )
    except (OSError, ValueError, KeyError, TypeError, NameError, RuntimeError) as error:
        result.update(
            status="TARGET_TEST_RANK_INPUTS_BLOCKED",
            blocker={"stage": stage, "type": type(error).__name__, "detail": str(error)},
            next_product_dependency="RESOLVE_RECORDED_TARGET_RANK_INPUT_GAP",
        )

    raw = json.dumps(result, ensure_ascii=False, sort_keys=True, indent=2,
                     allow_nan=False).encode("utf-8")
    digest = sha(raw)
    relative = reports / f"C149.R11_target_test_char_rrf_rank_inputs.{digest}.json"
    path = root / relative
    try:
        with path.open("xb") as handle:
            handle.write(raw)
            handle.flush()
            os.fsync(handle.fileno())
    except FileExistsError:
        pass
    require(path.read_bytes() == raw, "C149_REPORT_RELOAD_CONFLICT")
    result["output"] = {
        "path": relative.as_posix(),
        "file_bytes_sha256": digest,
        "reload_validation": "PASS",
    }
    details = {
        "hybrid100_order", "pool50_order", "local100_order",
        "local100_scores_by_chunk_id", "quota100_order",
        "char_channel_order", "char_rrf_order",
    }
    print(json.dumps(
        {key: value for key, value in result.items() if key not in details},
        ensure_ascii=False, indent=2,
    ))
    return result


V3_C149 = run_v3_c149()
````

````python
# C150 — R11.3 — Capture the target input receipt at the generation boundary
def run_v3_c150():
    def require(ok, code):
        if not ok:
            raise ValueError(code)

    def sha(raw):
        return hashlib.sha256(raw).hexdigest()

    def text_sha(text):
        return sha(text.encode("utf-8"))

    reports = Path("reports/phase2/quality_improvement/NB22E.lab.v3")
    cwd = Path.cwd()
    root = next((p.resolve() for p in (cwd, *cwd.parents)
                 if (p / reports).is_dir()), None)
    require(root is not None, "C150_PROJECT_ROOT_NOT_FOUND")

    def project_path(relative):
        relative = Path(relative)
        path = (root / relative).resolve()
        require(not relative.is_absolute() and path.is_relative_to(root),
                "C150_PATH_OUTSIDE_PROJECT")
        return path

    def read_bound(relative, digest):
        raw = project_path(relative).read_bytes()
        require(sha(raw) == digest, f"C150_HASH_CONFLICT:{relative}")
        return raw

    def save(stem, value):
        raw = json.dumps(
            value, ensure_ascii=False, sort_keys=True, indent=2, allow_nan=False,
        ).encode("utf-8")
        digest = sha(raw)
        relative = reports / f"{stem}.{digest}.json"
        path = project_path(relative)
        try:
            with path.open("xb") as handle:
                handle.write(raw)
                handle.flush()
                os.fsync(handle.fileno())
        except FileExistsError:
            pass
        require(path.read_bytes() == raw, "C150_RECEIPT_RELOAD_CONFLICT")
        return {"path": relative.as_posix(), "file_bytes_sha256": digest,
                "reload_validation": "PASS"}

    pins = {
        "C144": ("C144.R11_original_target_scenario_binding",
                 "edf453706d80836cb0cfd09b116e36bb7bad3169b1a372e23f0267fd29530570"),
        "C146": ("C146.R11_target_retrieval_input_binding",
                 "226aca0c92a8bd7f23f06b6714c7d427f65b46799a2b7fe6e9456cc7710eeb2e"),
        "C148": ("C148.R11_target_test_hybrid100_probe",
                 "8352cfa49bedc99c50335f57a66051a3613bdb1cf3618200dff771a8c1d654f4"),
        "C149": ("C149.R11_target_test_char_rrf_rank_inputs",
                 "46eee123bd73a40a0ad3c67e8a897c2abcaa5033aa0211831c740687cfefaa72"),
        "C139": ("C139.R11_persisted_opt_in_caller",
                 "3b5a43d5476dc28de79eed5fc35adc6e2a6834bbd9c6a341c8efa93b18669634"),
    }
    hashes = {key: digest for key, (_, digest) in pins.items()}
    result = {
        "cell": "C150",
        "roadmap_action": "R11",
        "roadmap_subtask": "R11.3",
        "scope": "ONE_TEST_TARGET_FIXED_RANKS_PRODUCT_PRE_GENERATION_BOUNDARY",
        "input_file_hashes": hashes,
        "target_pre_generation_receipt_verified": False,
        "product_runtime_binding_verified": False,
        "product_regression_tests_executed": 0,
        "responses_generated": 0,
        "new_api_calls": 0,
        "qdrant_queries_attempted": 0,
        "new_local_pairs_scored": 0,
        "done_item_count": 7,
        "open_item_count": 5,
        "newly_completed_items_from_previous_open_set": [],
        "selected_version_gate": "BLOCKED",
    }

    stage = "BIND_TARGET_AND_RANKS"
    events = []
    receipt_ref = None
    try:
        docs = {
            key: json.loads(read_bound(reports / f"{stem}.{digest}.json", digest))
            for key, (stem, digest) in pins.items()
        }
        spec, base, hybrid, rank, route = (
            docs[key] for key in ("C144", "C146", "C148", "C149", "C139")
        )
        keys = ("candidate_id", "question_sha256", "regression_spec_id")
        identity = tuple(spec[key] for key in keys)
        candidate, question_hash, spec_id = identity
        question = spec["question"]
        require(
            all(tuple(doc[key] for key in keys) == identity for doc in (base, hybrid, rank))
            and rank["status"] == "TARGET_TEST_CHAR_RRF_RANK_INPUTS_BUILT"
            and hybrid["status"] == "TARGET_TEST_HYBRID100_ORDER_MATCHED"
            and all(rank["input_file_hashes"][key] == hashes[key]
                    for key in ("C144", "C146", "C148"))
            and text_sha(question) == question_hash == rank["ranker_question_sha256"],
            "C150_TARGET_OR_RANK_RECEIPT_CONFLICT",
        )
        result.update(
            candidate_id=candidate,
            question_sha256=question_hash,
            regression_spec_id=spec_id,
            expected_behavior=spec["original_regression_spec"]["expected_behavior"],
            required_limit_disclosed=spec["original_regression_spec"]["required_limit_disclosed"],
        )

        hybrid_ids = hybrid["observed_hybrid100_order"]
        pool_ids = base["pool50_order"]
        pool_scores = base["pool50_scores_by_chunk_id"]
        top10 = rank["char_rrf_top10"]
        require(
            hybrid_ids == base["hybrid100_order"] == rank["hybrid100_order"]
            and len(hybrid_ids) == len(set(hybrid_ids)) == 100
            and pool_ids == rank["pool50_order"]
            and len(pool_ids) == len(set(pool_ids)) == 50
            and set(pool_ids) == set(hybrid_ids[:50]) == set(pool_scores)
            and all(rank["local100_scores_by_chunk_id"][cid] == score
                    for cid, score in pool_scores.items())
            and top10 == rank["char_rrf_order"][:10]
            and len(top10) == len(set(top10)) == 10,
            "C150_ORDER_OR_SCORE_BINDING_CONFLICT",
        )

        stage = "LOAD_PINNED_EVIDENCE_AND_PRODUCT_CODE"
        corpus_path = "data/phase2/chunks/nb11_benchmark_v1/baseline_phase1_char_safe_v1.jsonl"
        corpus_sha = rank["input_file_hashes"]["chunks"]
        wanted = set(top10)
        corpus = {}
        for line in read_bound(corpus_path, corpus_sha).splitlines():
            if line.strip():
                row = json.loads(line)
                cid = row["chunk_id"]
                if cid in wanted:
                    require(cid not in corpus, "C150_DUPLICATE_CHUNK")
                    corpus[cid] = row
        require(set(corpus) == wanted, "C150_REQUIRED_CHUNK_MISSING")
        chunks = [
            {"chunk_id": cid, "text": corpus[cid]["content"],
             "title": corpus[cid].get("title"), "url": corpus[cid].get("url")}
            for cid in top10
        ]
        text_hashes = [text_sha(chunk["text"]) for chunk in chunks]
        require(text_hashes == rank["char_rrf_top10_text_sha256"],
                "C150_CHUNK_TEXT_HASH_CONFLICT")

        require(route["status"] == "OPT_IN_CALLER_PERSISTED_IN_PRODUCT_SOURCE",
                "C150_C139_ROUTE_REQUIRED")
        route_ref = route["module"]
        route_scope = {}
        exec(compile(
            read_bound(route_ref["path"], route_ref["file_bytes_sha256"]),
            route_ref["path"], "exec", dont_inherit=True,
        ), route_scope)
        caller = route_scope[route_ref["export"]]

        pipeline_path = "src/pipeline/rag_pipeline.py"
        pipeline_sha = route["input_file_hashes"]["pipeline_source"]
        pipeline_source = read_bound(pipeline_path, pipeline_sha)
        pipeline_class = globals().get("RAGPipeline")
        require(isinstance(pipeline_class, type)
                and pipeline_class.__module__ == "src.pipeline.rag_pipeline",
                "C150_PRODUCT_CLASS_REQUIRED_IN_C001")
        code_type = type(run_v3_c150.__code__)
        compiled = compile(pipeline_source, pipeline_path, "exec", dont_inherit=True)
        class_code = next(
            code for code in compiled.co_consts
            if isinstance(code, code_type) and code.co_name == "RAGPipeline"
        )
        methods = {code.co_name: code for code in class_code.co_consts
                   if isinstance(code, code_type)}

        def signature(code):
            return (
                code.co_code,
                tuple(signature(value) if isinstance(value, code_type) else value
                      for value in code.co_consts),
                code.co_names, code.co_varnames, code.co_freevars, code.co_cellvars,
                code.co_flags, code.co_argcount, code.co_posonlyargcount,
                code.co_kwonlyargcount, code.co_exceptiontable,
            )

        for name in ("__init__", "ask_with_char_rrf", "build_context", "close"):
            require(signature(getattr(pipeline_class, name).__code__) == signature(methods[name]),
                    f"C150_LOADED_PRODUCT_METHOD_DIFFERS:{name}")
        hashes.update(chunks=corpus_sha, pipeline_source=pipeline_sha,
                      route_source=route_ref["file_bytes_sha256"])

        def query_check(question, candidate_id):
            require((candidate_id, text_sha(question)) == (candidate, question_hash),
                    "C150_PROVIDER_QUERY_CONFLICT")

        def hybrid_provider(*, question, candidate_id):
            query_check(question, candidate_id)
            events.append("fixed_hybrid100_supplied")
            return {
                "candidate_id": candidate, "question_sha256": question_hash,
                "ordered_chunk_ids": list(hybrid_ids), "receipt_sha256": hashes["C148"],
            }

        def pool_provider(*, question, candidate_id, hybrid100):
            query_check(question, candidate_id)
            require(list(hybrid100) == hybrid_ids, "C150_PROVIDER_HYBRID_CONFLICT")
            events.append("fixed_pool50_supplied")
            return {
                "candidate_id": candidate, "question_sha256": question_hash,
                "ordered_chunk_ids": list(pool_ids), "scores_by_chunk_id": dict(pool_scores),
                "receipt_sha256": hashes["C146"],
            }

        runtime_identity = {
            "scope": "ONE_TEST_TARGET_FIXED_RANKS_STOP_BEFORE_GENERATION",
            "pipeline_source_sha256": pipeline_sha,
            "route_source_sha256": route_ref["file_bytes_sha256"],
            "rank_input_receipt_sha256": hashes["C149"],
            "regression_spec_id": spec_id,
        }

        def rank_provider(*, question, candidate_id, top_k, hybrid100,
                          pool50_order, pool50_scores):
            query_check(question, candidate_id)
            require(top_k == 10 and list(hybrid100) == hybrid_ids
                    and list(pool50_order) == pool_ids and pool50_scores == pool_scores_ref,
                    "C150_FINAL_RANK_INPUT_CONFLICT")
            events.append("fixed_char_rrf_supplied")
            return {
                "candidate_id": candidate, "question_sha256": question_hash,
                "selected_variant": "char_rrf",
                "runtime_identity": dict(runtime_identity),
                "ordered_chunks": chunks,
            }

        pool_scores_ref = pool_scores

        def write_receipt(role, payload):
            nonlocal receipt_ref
            require(role == "input" and receipt_ref is None, "C150_UNEXPECTED_RECEIPT_WRITE")
            receipt_ref = save("C150.R11_target_product_input_receipt", payload)
            events.append("input_receipt_written")
            return receipt_ref["file_bytes_sha256"]

        class StopBeforeGeneration(Exception):
            pass

        class GenerationBoundary:
            def generate_answer(self, *, question, context):
                query_check(question, candidate)
                require(receipt_ref is not None, "C150_INPUT_RECEIPT_REQUIRED")
                recorded = json.loads(read_bound(
                    receipt_ref["path"], receipt_ref["file_bytes_sha256"],
                ))
                require(
                    recorded["candidate_id"] == candidate
                    and recorded["question_sha256"] == question_hash
                    and recorded["selected_variant"] == "char_rrf"
                    and recorded["ordered_chunk_ids"] == top10
                    and recorded["ordered_chunk_text_sha256"] == text_hashes
                    and recorded["rendered_context_sha256"] == text_sha(context)
                    and recorded["runtime_identity"] == {
                        **runtime_identity,
                        "hybrid100_provider_receipt_sha256": hashes["C148"],
                        "pool50_provider_receipt_sha256": hashes["C146"],
                    },
                    "C150_PRE_GENERATION_RECEIPT_CONFLICT",
                )
                events.append("input_receipt_verified_at_generation_boundary")
                raise StopBeforeGeneration

        stage = "RUN_PRODUCT_TO_GENERATION_BOUNDARY"
        storage = project_path(hybrid["dense_storage_path"])
        require(storage.is_dir(), "C150_LOCAL_DENSE_STORAGE_REQUIRED")
        pipeline = pipeline_class(
            vectorstore_path=storage,
            collection_name=hybrid["dense_collection"],
            use_cache=False,
        )
        try:
            pipeline.generator = GenerationBoundary()
            try:
                caller(
                    pipeline, question=question, candidate_id=candidate,
                    hybrid100_provider=hybrid_provider, pool50_provider=pool_provider,
                    rank_provider=rank_provider, write_receipt=write_receipt, top_k=10,
                )
            except StopBeforeGeneration:
                pass
            else:
                raise ValueError("C150_EXPECTED_GENERATION_BOUNDARY_STOP")
        finally:
            pipeline.close()

        require(events == [
            "fixed_hybrid100_supplied", "fixed_pool50_supplied",
            "fixed_char_rrf_supplied", "input_receipt_written",
            "input_receipt_verified_at_generation_boundary",
        ], "C150_EVENT_ORDER_CONFLICT")
        result.update(
            status="TARGET_FIXED_RANK_PRODUCT_INPUT_RECEIPT_VERIFIED",
            blocker=None,
            target_pre_generation_receipt_verified=True,
            next_product_dependency="BIND_GENERATION_CONFIGURATION_AND_ASSESS_TARGET_RESPONSE",
        )
    except (OSError, ValueError, KeyError, TypeError, RuntimeError,
            AttributeError, ImportError, StopIteration) as error:
        result.update(
            status="TARGET_PRODUCT_INPUT_RECEIPT_BLOCKED",
            blocker={"stage": stage, "type": type(error).__name__, "detail": str(error)},
            next_product_dependency="RESOLVE_RECORDED_TARGET_RECEIPT_BLOCKER",
        )

    result.update(events=events, input_receipt=receipt_ref)
    result["output"] = save("C150.R11_target_pre_generation_product_probe", result)
    print(json.dumps(result, ensure_ascii=False, indent=2))
    return result


V3_C150 = run_v3_c150()
````

````python
# C153 — R11.3 — Check whether the earlier DeepSeek batch has the exact target request
def run_v3_c153():
    def require(condition, code):
        if not condition:
            raise ValueError(code)

    def sha(raw):
        return hashlib.sha256(raw).hexdigest()

    reports = Path("reports/phase2/quality_improvement/NB22E.lab.v3")
    cwd = Path.cwd()
    root = next(
        (path.resolve() for path in (cwd, *cwd.parents) if (path / reports).is_dir()),
        None,
    )
    require(root is not None, "C153_PROJECT_ROOT_NOT_FOUND")

    def read_bound(relative, digest):
        relative = Path(relative)
        path = (root / relative).resolve()
        require(
            not relative.is_absolute() and path.is_relative_to(root),
            "C153_PATH_OUTSIDE_PROJECT",
        )
        raw = path.read_bytes()
        require(sha(raw) == digest, f"C153_HASH_CONFLICT:{relative}")
        return json.loads(raw)

    pins = {
        "C152": (
            "C152.R11_target_offline_generation_profile",
            "3fc635ddb6b435cec096351b7a9db8b97557cf4dd0d7c27c28e4afafe3c6bcbe",
        ),
        "C042": (
            "C042.R11_deepseek_batch",
            "e8ff5073b9a0c595dda7a0cd7d49975518dac17c03c9130b9ae3c3cd35b60e8b",
        ),
        "C043": (
            "C043.R11_scenario_review_inputs",
            "811af55fabd63d695bd2767c762055c4997d81e4acd7c36f7bdde717421e6145",
        ),
    }
    docs = {
        cell: read_bound(reports / f"{stem}.{digest}.json", digest)
        for cell, (stem, digest) in pins.items()
    }

    prior = docs["C152"]
    require(
        prior["status"] == "TARGET_GENERATOR_REQUEST_CAPTURED_OFFLINE"
        and prior["offline_request_profile_verified"] is True,
        "C153_SUCCESSFUL_C152_REQUIRED",
    )
    ref = prior["request_receipt"]
    captured = read_bound(ref["path"], ref["file_bytes_sha256"])
    identity = ("candidate_id", "question_sha256", "regression_spec_id")
    require(
        all(captured[key] == prior[key] for key in identity)
        and captured["dispatch_performed"] is False,
        "C153_C152_REQUEST_IDENTITY_CONFLICT",
    )

    expected_messages = captured["request"]["messages"]
    require(
        [message["role"] for message in expected_messages] == prior["message_roles"]
        and [sha(message["content"].encode("utf-8")) for message in expected_messages]
            == prior["message_text_sha256"],
        "C153_C152_MESSAGE_RECEIPT_CONFLICT",
    )
    candidate = prior["candidate_id"]

    def escape(key):
        return str(key).replace("~", "~0").replace("/", "~1")

    def walk(value, pointer="", owner=None, owner_pointer=None):
        if isinstance(value, dict):
            # A nested explicit candidate ID overrides an outer one.
            if isinstance(value.get("candidate_id"), str):
                owner, owner_pointer = value["candidate_id"], pointer
            yield pointer, value, owner, owner_pointer
            for key, child in value.items():
                yield from walk(child, f"{pointer}/{escape(key)}", owner, owner_pointer)
        elif isinstance(value, list):
            for index, child in enumerate(value):
                yield from walk(child, f"{pointer}/{index}", owner, owner_pointer)

    evidence = {}
    exact_matches = []
    for cell in ("C042", "C043"):
        target_records = set()
        message_records = []
        for pointer, row, owner, owner_pointer in walk(docs[cell]):
            if owner != candidate:
                continue
            target_records.add(owner_pointer)
            messages = row.get("messages")
            if not isinstance(messages, list) or not messages:
                continue
            if not all(
                isinstance(message, dict)
                and isinstance(message.get("role"), str)
                and isinstance(message.get("content"), str)
                for message in messages
            ):
                continue

            model = row.get("model")
            exact = messages == expected_messages
            observation = {
                "source": cell,
                "candidate_pointer": owner_pointer,
                "request_pointer": pointer,
                "model_in_request_record": model if isinstance(model, str) else None,
                "exact_ordered_messages_match": exact,
                "message_text_sha256": [
                    sha(message["content"].encode("utf-8")) for message in messages
                ],
            }
            message_records.append(observation)
            if exact and model == "DeepSeek-V4-Flash":
                exact_matches.append(observation)

        evidence[cell] = {
            "target_record_pointers": sorted(target_records),
            "inline_request_message_records": message_records,
        }

    request_records = sum(
        len(item["inline_request_message_records"]) for item in evidence.values()
    )
    if exact_matches:
        status = "EXACT_DEEPSEEK_REQUEST_MESSAGES_FOUND_RESPONSE_BINDING_PENDING"
        next_step = "VERIFY_LINKED_DEEPSEEK_RESPONSE_AND_EXECUTION_PARAMETERS"
    elif request_records:
        status = "EXACT_DEEPSEEK_REQUEST_NOT_ESTABLISHED_IN_INLINE_RECORDS"
        next_step = "RESOLVE_TARGET_REQUEST_RECEIPT_OR_DOCUMENT_MESSAGE_DIFFERENCE"
    else:
        status = "TARGET_REQUEST_MESSAGES_NOT_FOUND_INLINE"
        next_step = "FOLLOW_TARGET_C042_REQUEST_AND_RESPONSE_RECEIPTS"

    result = {
        "cell": "C153",
        "roadmap_action": "R11",
        "roadmap_subtask": "R11.3",
        "status": status,
        **{key: prior[key] for key in identity},
        "input_file_hashes": {
            **{key: digest for key, (_, digest) in pins.items()},
            "C152_request": ref["file_bytes_sha256"],
        },
        "scope": "EXPLICIT_CANDIDATE_AND_INLINE_REQUEST_RECORDS_IN_C042_C043",
        "continuation_model": "DeepSeek-V4-Flash",
        "c152_capture_model": captured["request"]["model"],
        "candidate_evidence": evidence,
        "exact_deepseek_request_matches": exact_matches,
        "prior_response_reused": False,
        "product_regression_tests_executed": 0,
        "new_api_calls": 0,
        "done_item_count": 7,
        "open_item_count": 5,
        "selected_version_gate": "BLOCKED",
        "next_product_dependency": next_step,
    }

    raw = json.dumps(
        result, ensure_ascii=False, sort_keys=True, indent=2, allow_nan=False,
    ).encode("utf-8")
    digest = sha(raw)
    relative = reports / f"C153.R11_existing_deepseek_request_check.{digest}.json"
    path = root / relative
    try:
        with path.open("xb") as handle:
            handle.write(raw)
            handle.flush()
            os.fsync(handle.fileno())
    except FileExistsError:
        pass
    require(path.read_bytes() == raw, "C153_REPORT_RELOAD_CONFLICT")
    result["output"] = {
        "path": relative.as_posix(),
        "file_bytes_sha256": digest,
        "reload_validation": "PASS",
    }
    print(json.dumps(result, ensure_ascii=False, indent=2))
    return result


V3_C153 = run_v3_c153()
````

````python
# C154 — R11.3 — Bind the prior DeepSeek request and response offline
def run_v3_c154():
    def require(ok, code):
        if not ok:
            raise ValueError(code)

    def sha(raw):
        return hashlib.sha256(raw).hexdigest()

    def encode(value):
        return json.dumps(
            value, ensure_ascii=False, sort_keys=True,
            separators=(",", ":"), allow_nan=False,
        ).encode("utf-8")

    reports = Path("reports/phase2/quality_improvement/NB22E.lab.v3")
    cwd = Path.cwd()
    root = next(
        (p.resolve() for p in (cwd, *cwd.parents) if (p / reports).is_dir()),
        None,
    )
    require(root is not None, "C154_PROJECT_ROOT_NOT_FOUND")

    def path_for(relative):
        relative = Path(relative)
        path = (root / relative).resolve()
        require(not relative.is_absolute() and path.is_relative_to(root),
                "C154_PATH_OUTSIDE_PROJECT")
        return path

    def load(relative, digest):
        raw = path_for(relative).read_bytes()
        require(sha(raw) == digest, f"C154_HASH_CONFLICT:{relative}")
        return json.loads(raw)

    def save(stem, value):
        raw = encode(value)
        digest = sha(raw)
        relative = reports / f"{stem}.{digest}.json"
        path = path_for(relative)
        try:
            with path.open("xb") as handle:
                handle.write(raw)
                handle.flush()
                os.fsync(handle.fileno())
        except FileExistsError:
            pass
        require(path.read_bytes() == raw, "C154_REPORT_RELOAD_CONFLICT")
        return {"path": relative.as_posix(), "file_bytes_sha256": digest,
                "reload_validation": "PASS"}

    pins = {
        "C153": (
            "C153.R11_existing_deepseek_request_check."
            "86f345e939fd5290fc44913e674111d0d0f5602da33e05827d5922a81f02bd8f.json",
            "86f345e939fd5290fc44913e674111d0d0f5602da33e05827d5922a81f02bd8f",
        ),
        "C040_plan": (
            "C040.R11_generation_plan."
            "608d2551794d2028f973e23a89e96079b0f8b2fead857f43c16079379f29b3bd.json",
            "cda83957141dcfbf85e2073b9af21e65d8406da2e328da6c2b5f25a713579e85",
        ),
    }
    result = {
        "cell": "C154",
        "roadmap_action": "R11",
        "roadmap_subtask": "R11.3",
        "input_file_hashes": {key: digest for key, (_, digest) in pins.items()},
        "prior_response_bytes_verified": False,
        "prior_request_id_reproduced": None,
        "exact_ordered_messages_match": None,
        "prior_response_reused_for_selected_regression": False,
        "product_regression_tests_executed": 0,
        "new_api_calls": 0,
        "done_item_count": 7,
        "open_item_count": 5,
        "selected_version_gate": "BLOCKED",
    }
    stage = "BIND_TARGET_AND_PINNED_REPORTS"
    try:
        prior = load(reports / pins["C153"][0], pins["C153"][1])
        require(
            prior["status"] == "TARGET_REQUEST_MESSAGES_NOT_FOUND_INLINE"
            and prior["candidate_evidence"]["C042"]["target_record_pointers"] == ["/receipts/1"]
            and "/responses/1" in prior["candidate_evidence"]["C043"]["target_record_pointers"],
            "C154_C153_TARGET_POINTER_CONFLICT",
        )
        candidate = prior["candidate_id"]
        spec_id = prior["regression_spec_id"]
        result.update(candidate_id=candidate, regression_spec_id=spec_id,
                      question_sha256=prior["question_sha256"])
        hashes = prior["input_file_hashes"]

        def report(stem, key):
            digest = hashes[key]
            document = load(reports / f"{stem}.{digest}.json", digest)
            result["input_file_hashes"][key] = digest
            return document

        batch = report("C042.R11_deepseek_batch", "C042")
        review = report("C043.R11_scenario_review_inputs", "C043")
        current = report("C152.R11_target_offline_generation_request", "C152_request")
        batch_row = batch["receipts"][1]
        review_row = review["responses"][1]
        require(
            batch_row["candidate_id"] == review_row["candidate_id"]
                == current["candidate_id"] == candidate
            and current["question_sha256"] == prior["question_sha256"]
            and current["regression_spec_id"] == spec_id,
            "C154_TARGET_IDENTITY_CONFLICT",
        )

        stage = "VERIFY_SAVED_DEEPSEEK_RESPONSE"
        request_id = review_row["request_id"]
        require(
            isinstance(request_id, str) and len(request_id) == 64
            and all(c in "0123456789abcdef" for c in request_id),
            "C154_INVALID_REQUEST_ID",
        )
        receipt_dir = reports / "C042.R11_deepseek_receipts"
        response_path = receipt_dir / f"{request_id}.response.json"
        require(
            batch_row["receipt_path"] == review_row["receipt_path"] == response_path.as_posix(),
            "C154_RESPONSE_RECEIPT_PATH_CONFLICT",
        )
        response = load(response_path, review_row["receipt_file_bytes_sha256"])
        started = json.loads(
            path_for(receipt_dir / f"{request_id}.started.json").read_bytes()
        )
        require(
            response["status"] == "HTTP_200_RECEIVED"
            and response["request_id"] == started["request_id"] == request_id
            and response["candidate_id"] == started["candidate_id"] == candidate,
            "C154_STARTED_OR_RESPONSE_IDENTITY_CONFLICT",
        )
        body = response["response_body"].encode("utf-8")
        body_sha = sha(body)
        require(
            body_sha == response["response_bytes_sha256"]
                == batch_row["response_bytes_sha256"] == review_row["response_bytes_sha256"],
            "C154_RESPONSE_BODY_HASH_CONFLICT",
        )
        envelope = json.loads(body)
        choice = envelope["choices"][0]
        answer = choice["message"]["content"]
        model = envelope["model"].strip().rsplit("/", 1)[-1].lower()
        require(
            isinstance(answer, str) and bool(answer.strip())
            and answer == review_row["response_content"]
            and model == batch_row["response_model"]
            and model in {"deepseek-v4-flash", "deepseek-v4-flash-0731"},
            "C154_RESPONSE_CONTENT_OR_MODEL_CONFLICT",
        )
        result.update(
            prior_response_bytes_verified=True,
            prior_response_model=model,
            prior_response_finish_reason=choice.get("finish_reason"),
            prior_response_text_sha256=sha(answer.encode("utf-8")),
            prior_response_receipt_path=response_path.as_posix(),
        )

        stage = "REBUILD_C042_REQUEST_FROM_PINNED_PLAN"
        plan = load(reports / pins["C040_plan"][0], pins["C040_plan"][1])
        jobs = [job for job in plan["jobs"] if job["candidate_id"] == candidate]
        require(len(jobs) == 1, "C154_EXPECTED_ONE_PLAN_JOB")
        job = jobs[0]
        require(job["job_sha256"] == review_row["source_plan_job_sha256"],
                "C154_PLAN_JOB_RECEIPT_CONFLICT")
        historical_messages = job["request"]["messages"]
        current_messages = current["request"]["messages"]
        for messages in (historical_messages, current_messages):
            require(
                isinstance(messages, list) and bool(messages)
                and all(isinstance(m, dict) and isinstance(m.get("role"), str)
                        and isinstance(m.get("content"), str) for m in messages),
                "C154_INVALID_MESSAGE_LIST",
            )

        payload = {
            "model": "DeepSeek-V4-Flash",
            "messages": historical_messages,
            "temperature": 0,
            "max_tokens": 600,
            "thinking": {"type": "disabled"},
        }
        payload_sha = sha(encode(payload))
        result.update(
            prior_request_payload_sha256=payload_sha,
            exact_ordered_messages_match=(historical_messages == current_messages),
            message_comparison=[
                {
                    "index": i,
                    "prior_role": old["role"] if old is not None else None,
                    "current_role": new["role"] if new is not None else None,
                    "exact_match": old == new,
                    "prior_text_sha256": sha(old["content"].encode("utf-8"))
                        if old is not None else None,
                    "current_text_sha256": sha(new["content"].encode("utf-8"))
                        if new is not None else None,
                }
                for i in range(max(len(historical_messages), len(current_messages)))
                for old, new in [(
                    historical_messages[i] if i < len(historical_messages) else None,
                    current_messages[i] if i < len(current_messages) else None,
                )]
            ],
        )

        stage = "REPRODUCE_C042_REQUEST_ID_WITH_LOCAL_GATEWAY"
        parser = globals().get("dotenv_values")
        url_module = globals().get("urllib")
        if not callable(parser) or not hasattr(url_module, "parse"):
            result["request_id_binding_gap"] = "C042_DOTENV_OR_URL_PARSER_NOT_IN_KERNEL"
        elif not (root / ".env").is_file():
            result["request_id_binding_gap"] = "PROJECT_DOTENV_NOT_AVAILABLE"
        else:
            config = parser(root / ".env", interpolate=False)
            base = str(config.get("ARVAN_GATEWAY_URL") or "").strip()
            try:
                url = url_module.parse.urlsplit(base)
                valid = (
                    url.scheme == "https" and bool(url.hostname)
                    and url.username is None and url.password is None
                    and not url.query and not url.fragment
                )
            except ValueError:
                valid = False
            if valid:
                route = url.path.rstrip("/")
                if not route:
                    route = "/v1/chat/completions"
                elif not route.endswith("/chat/completions"):
                    route += "/chat/completions"
                endpoint_sha = sha(url._replace(path=route).geturl().encode("utf-8"))
                rebuilt_id = sha(encode({
                    "candidate_id": candidate,
                    "plan_sha256": pins["C040_plan"][1],
                    "endpoint_sha256": endpoint_sha,
                    "request_sha256": payload_sha,
                }))
                result["prior_request_id_reproduced"] = rebuilt_id == request_id
                if rebuilt_id != request_id:
                    result["request_id_binding_gap"] = "REQUEST_ID_DIFFERS_WITH_CURRENT_GATEWAY"
            else:
                result["request_id_binding_gap"] = "CURRENT_GATEWAY_URL_NOT_VALIDATED"

        if result["prior_request_id_reproduced"] is not True:
            status = "PRIOR_RESPONSE_VERIFIED_REQUEST_ID_BINDING_OPEN"
            next_step = "RESOLVE_HISTORICAL_REQUEST_ID_BINDING"
        elif result["exact_ordered_messages_match"]:
            status = "PRIOR_DEEPSEEK_REQUEST_AND_RESPONSE_BOUND_MESSAGES_MATCH"
            next_step = "CHECK_GENERATION_PARAMETERS_AND_EVIDENCE_SCOPE_BEFORE_REUSE"
        else:
            status = "PRIOR_DEEPSEEK_REQUEST_AND_RESPONSE_BOUND_MESSAGES_DIFFER"
            next_step = "DETERMINE_WHICH_PROMPT_OR_CONTEXT_COMPONENT_DIFFERS"
        result.update(status=status, blocker=None, next_product_dependency=next_step)

    except (OSError, ValueError, KeyError, TypeError, IndexError) as error:
        result.update(
            status="PRIOR_DEEPSEEK_BINDING_BLOCKED",
            blocker={"stage": stage, "type": type(error).__name__, "detail": str(error)},
            next_product_dependency="RESOLVE_RECORDED_RECEIPT_BINDING_GAP",
        )

    result["output"] = save("C154.R11_prior_deepseek_request_response_binding", result)
    print(json.dumps(result, ensure_ascii=False, indent=2))
    return result


V3_C154 = run_v3_c154()
````

### سل ۱۵۵ — تعیین منشأ اختلاف پیام کاربر

````python
# C155 — R11.3 — Identify the user-message difference without resending
def run_v3_c155():
    def require(ok, code):
        if not ok:
            raise ValueError(code)

    def sha(raw):
        return hashlib.sha256(raw).hexdigest()

    def text_sha(text):
        return sha(text.encode("utf-8"))

    def encode(value):
        return json.dumps(
            value, ensure_ascii=False, sort_keys=True,
            separators=(",", ":"), allow_nan=False,
        ).encode("utf-8")

    reports = Path("reports/phase2/quality_improvement/NB22E.lab.v3")
    cwd = Path.cwd()
    root = next(
        (p.resolve() for p in (cwd, *cwd.parents) if (p / reports).is_dir()),
        None,
    )
    require(root is not None, "C155_PROJECT_ROOT_NOT_FOUND")

    def project_path(relative):
        relative = Path(relative)
        path = (root / relative).resolve()
        require(not relative.is_absolute() and path.is_relative_to(root),
                "C155_PATH_OUTSIDE_PROJECT")
        return path

    def load(relative, digest):
        raw = project_path(relative).read_bytes()
        require(sha(raw) == digest, f"C155_HASH_CONFLICT:{relative}")
        return json.loads(raw)

    def load_ref(ref):
        return load(ref["path"], ref["file_bytes_sha256"])

    pins = {
        "C154": (
            "C154.R11_prior_deepseek_request_response_binding",
            "4c46b189e0649208455f099e1274c793a8fa333f57ccae1926785e3c238358ac",
        ),
        "C151": (
            "C151.R11_target_generation_configuration_binding",
            "18e5feafd6c8dafceb754346bd0d22afadda5e42d10f33d80fc81459df903045",
        ),
    }

    def report(stem, digest):
        return load(reports / f"{stem}.{digest}.json", digest)

    old_run = report(*pins["C154"])
    current_input = report(*pins["C151"])
    require(
        old_run["status"] == "PRIOR_DEEPSEEK_REQUEST_AND_RESPONSE_BOUND_MESSAGES_DIFFER"
        and old_run["prior_response_bytes_verified"] is True
        and old_run["prior_request_id_reproduced"] is True
        and current_input["exact_context_verified"] is True,
        "C155_PRIOR_EVIDENCE_REQUIRED",
    )

    keys = ("candidate_id", "question_sha256", "regression_spec_id")
    identity = {key: old_run[key] for key in keys}
    prepared = load_ref(current_input["prepared_input"])
    hashes = old_run["input_file_hashes"]
    current = report("C152.R11_target_offline_generation_request", hashes["C152_request"])
    require(
        all(current_input[key] == prepared[key] == current[key] == identity[key]
            for key in keys),
        "C155_TARGET_IDENTITY_CONFLICT",
    )

    plan = load(
        reports / (
            "C040.R11_generation_plan."
            "608d2551794d2028f973e23a89e96079b0f8b2fead857f43c16079379f29b3bd.json"
        ),
        hashes["C040_plan"],
    )
    jobs = [job for job in plan["jobs"] if job["candidate_id"] == identity["candidate_id"]]
    require(len(jobs) == 1, "C155_EXPECTED_ONE_PLAN_JOB")
    previous_messages = jobs[0]["request"]["messages"]
    current_messages = current["request"]["messages"]
    payload = {
        "model": "DeepSeek-V4-Flash",
        "messages": previous_messages,
        "temperature": 0,
        "max_tokens": 600,
        "thinking": {"type": "disabled"},
    }
    require(
        sha(encode(payload)) == old_run["prior_request_payload_sha256"],
        "C155_PREVIOUS_REQUEST_HASH_CONFLICT",
    )
    require(
        [message["role"] for message in previous_messages]
        == [message["role"] for message in current_messages] == ["system", "user"]
        and previous_messages[0] == current_messages[0],
        "C155_SYSTEM_OR_MESSAGE_STRUCTURE_CONFLICT",
    )

    before = previous_messages[1]["content"]
    after = current_messages[1]["content"]
    comparison = old_run["message_comparison"][1]
    require(
        text_sha(before) == comparison["prior_text_sha256"]
        and text_sha(after) == comparison["current_text_sha256"]
        and before != after,
        "C155_USER_MESSAGE_HASH_CONFLICT",
    )

    question, context = prepared["question"], prepared["context"]
    require(
        bool(question.strip()) and bool(context)
        and text_sha(question) == identity["question_sha256"]
        and text_sha(context) == prepared["rendered_context_sha256"]
            == current["rendered_context_sha256"]
            == current_input["rendered_context_sha256"],
        "C155_QUESTION_OR_CONTEXT_HASH_CONFLICT",
    )
    require(after.count(context) == 1, "C155_CURRENT_CONTEXT_BOUNDARY_AMBIGUOUS")
    prefix, _, suffix = after.partition(context)
    require(bool(prefix or suffix), "C155_CURRENT_WRAPPER_EMPTY")

    context_count = before.count(context)
    same_wrapper = (
        before.startswith(prefix) and before.endswith(suffix)
        and len(before) >= len(prefix) + len(suffix)
    )
    old_context = None
    if same_wrapper:
        end = len(before) - len(suffix)
        old_context = before[len(prefix):end]
        status = (
            "CONTEXT_DIFFERS_WITH_SAME_USER_WRAPPER"
            if old_context != context else "USER_MESSAGE_DIFFERENCE_UNRESOLVED"
        )
    elif context_count == 1:
        old_context = context
        status = "USER_WRAPPER_DIFFERS_WITH_EXACT_CONTEXT_PRESENT"
    else:
        status = "HISTORICAL_CONTEXT_BOUNDARIES_UNRESOLVED"

    result = {
        "cell": "C155",
        "roadmap_action": "R11",
        "roadmap_subtask": "R11.3",
        "status": status,
        **identity,
        "input_file_hashes": {
            **{key: digest for key, (_, digest) in pins.items()},
            "C040_plan": hashes["C040_plan"],
            "C152_request": hashes["C152_request"],
            "C151_prepared_input": current_input["prepared_input"]["file_bytes_sha256"],
        },
        "system_message_exact_match": True,
        "current_question_occurs_in_prior_user_message": question in before,
        "current_question_occurs_in_current_user_message": question in after,
        "current_question_occurs_in_current_wrapper": question in prefix + suffix,
        "exact_current_context_occurrences_in_prior_user_message": context_count,
        "current_wrapper_matches_prior_boundaries": same_wrapper,
        "prior_context_sha256": text_sha(old_context) if old_context is not None else None,
        "current_context_sha256": text_sha(context),
        "prior_context_exact_match": (
            old_context == context if old_context is not None else None
        ),
        "prior_response_preserved_as_diagnostic": True,
        "prior_response_reused_for_selected_regression": False,
        "new_api_calls": 0,
        "product_regression_tests_executed": 0,
        "done_item_count": 7,
        "open_item_count": 5,
        "selected_version_gate": "BLOCKED",
        "next_product_dependency": (
            "PREPARE_DEEPSEEK_REQUEST_FOR_CURRENT_BOUND_INPUT"
            if status in {
                "CONTEXT_DIFFERS_WITH_SAME_USER_WRAPPER",
                "USER_WRAPPER_DIFFERS_WITH_EXACT_CONTEXT_PRESENT",
            }
            else "BIND_HISTORICAL_USER_MESSAGE_TO_C039_EVIDENCE_BOUNDARY"
        ),
    }

    raw = encode(result)
    digest = sha(raw)
    relative = reports / f"C155.R11_deepseek_user_message_difference.{digest}.json"
    path = project_path(relative)
    try:
        with path.open("xb") as handle:
            handle.write(raw)
            handle.flush()
            os.fsync(handle.fileno())
    except FileExistsError:
        pass
    require(path.read_bytes() == raw, "C155_REPORT_RELOAD_CONFLICT")
    result["output"] = {
        "path": relative.as_posix(),
        "file_bytes_sha256": digest,
        "reload_validation": "PASS",
    }
    print(json.dumps(result, ensure_ascii=False, indent=2))
    return result


V3_C155 = run_v3_c155()
````

````python
# C156 — R11.3 — Prepare one exact DeepSeek request without dispatch
def run_v3_c156():
    def require(ok, code):
        if not ok:
            raise ValueError(code)

    def sha(raw):
        return hashlib.sha256(raw).hexdigest()

    def encode(value):
        return json.dumps(
            value, ensure_ascii=False, sort_keys=True,
            separators=(",", ":"), allow_nan=False,
        ).encode("utf-8")

    reports = Path("reports/phase2/quality_improvement/NB22E.lab.v3")
    cwd = Path.cwd()
    root = next(
        (p.resolve() for p in (cwd, *cwd.parents) if (p / reports).is_dir()),
        None,
    )
    require(root is not None, "C156_PROJECT_ROOT_NOT_FOUND")

    def path_for(relative):
        relative = Path(relative)
        path = (root / relative).resolve()
        require(
            not relative.is_absolute() and path.is_relative_to(root),
            "C156_PATH_OUTSIDE_PROJECT",
        )
        return path

    def load(relative, digest):
        raw = path_for(relative).read_bytes()
        require(sha(raw) == digest, f"C156_HASH_CONFLICT:{relative}")
        return json.loads(raw)

    def report(stem, digest):
        return load(reports / f"{stem}.{digest}.json", digest)

    def save(stem, payload):
        raw = encode(payload)
        digest = sha(raw)
        relative = reports / f"{stem}.{digest}.json"
        path = path_for(relative)
        try:
            with path.open("xb") as handle:
                handle.write(raw)
                handle.flush()
                os.fsync(handle.fileno())
        except FileExistsError:
            pass
        require(path.read_bytes() == raw, "C156_REPORT_RELOAD_CONFLICT")
        return {
            "path": relative.as_posix(),
            "file_bytes_sha256": digest,
            "reload_validation": "PASS",
        }

    c155_sha = "7fac0ad2c8aabece01f075ab2d4b4a5cd926303d452f1a1ac6427d436651fd46"
    comparison = report("C155.R11_deepseek_user_message_difference", c155_sha)
    require(
        comparison["status"] == "CONTEXT_DIFFERS_WITH_SAME_USER_WRAPPER"
        and comparison["system_message_exact_match"] is True
        and comparison["current_wrapper_matches_prior_boundaries"] is True
        and comparison["prior_context_exact_match"] is False,
        "C156_CONTEXT_DIFFERENCE_EVIDENCE_REQUIRED",
    )

    hashes = comparison["input_file_hashes"]
    previous = report("C154.R11_prior_deepseek_request_response_binding", hashes["C154"])
    prepared = report("C151.R11_target_generation_input", hashes["C151_prepared_input"])
    capture = report("C152.R11_target_offline_generation_request", hashes["C152_request"])
    identity = {
        key: comparison[key]
        for key in ("candidate_id", "question_sha256", "regression_spec_id")
    }
    require(
        all(document[key] == value
            for document in (previous, prepared, capture)
            for key, value in identity.items())
        and previous["prior_request_id_reproduced"] is True
        and previous["prior_response_bytes_verified"] is True
        and capture["dispatch_performed"] is False,
        "C156_TARGET_OR_PRIOR_REQUEST_BINDING_CONFLICT",
    )

    question = prepared["question"]
    context = prepared["context"]
    context_sha = sha(context.encode("utf-8"))
    input_ref = prepared["product_input_receipt"]
    input_receipt = load(input_ref["path"], input_ref["file_bytes_sha256"])
    require(
        sha(question.encode("utf-8")) == identity["question_sha256"]
        and context_sha == comparison["current_context_sha256"]
            == prepared["rendered_context_sha256"]
            == capture["rendered_context_sha256"]
            == input_receipt["rendered_context_sha256"]
        and input_ref["file_bytes_sha256"] == capture["input_file_hashes"]["C150_input"]
        and input_receipt["candidate_id"] == identity["candidate_id"]
        and input_receipt["question_sha256"] == identity["question_sha256"]
        and input_receipt["selected_variant"] == "char_rrf"
        and input_receipt["runtime_identity"]["regression_spec_id"]
            == identity["regression_spec_id"],
        "C156_PRE_GENERATION_RECEIPT_CONFLICT",
    )

    messages = capture["request"]["messages"]
    require(
        [message["role"] for message in messages] == ["system", "user"]
        and [sha(message["content"].encode("utf-8")) for message in messages]
            == [row["current_text_sha256"] for row in previous["message_comparison"]]
        and messages[1]["content"].count(context) == 1
        and question in messages[1]["content"],
        "C156_EXACT_MESSAGE_BINDING_CONFLICT",
    )

    profile = {
        "model": "DeepSeek-V4-Flash",
        "temperature": 0,
        "max_tokens": 600,
        "thinking": {"type": "disabled"},
    }
    request = {**profile, "messages": messages}
    request_sha = sha(encode(request))
    require(
        request_sha != previous["prior_request_payload_sha256"],
        "C156_REQUEST_UNEXPECTEDLY_MATCHES_PRIOR_PAYLOAD",
    )

    request_ref = save("C156.R11_target_deepseek_request", {
        "scope": "ONE_TEST_TARGET_FIXED_RANKS_DEEPSEEK_REQUEST_PREPARATION",
        **identity,
        "input_file_hashes": {
            "C155": c155_sha,
            "C154": hashes["C154"],
            "C152_request": hashes["C152_request"],
            "C151_prepared_input": hashes["C151_prepared_input"],
            "C150_input": input_ref["file_bytes_sha256"],
        },
        "rendered_context_sha256": context_sha,
        "ordered_chunk_ids": input_receipt["ordered_chunk_ids"],
        "request_profile_basis": "REUSE_C042_REQUEST_SETTINGS_WITH_CURRENT_BOUND_MESSAGES",
        "request_payload_sha256": request_sha,
        "request": request,
        "dispatch_performed": False,
    })

    result = {
        "cell": "C156",
        "roadmap_action": "R11",
        "roadmap_subtask": "R11.3",
        "status": "TARGET_DEEPSEEK_REQUEST_PREPARED_OFFLINE",
        **identity,
        "C155_sha256": c155_sha,
        "request_receipt": request_ref,
        "request_payload_sha256": request_sha,
        "prior_request_payload_sha256": previous["prior_request_payload_sha256"],
        "rendered_context_sha256": context_sha,
        "prepared_profile": profile,
        "message_roles": [message["role"] for message in messages],
        "request_bytes": len(encode(request)),
        "current_rate_verified": False,
        "estimated_cost_toman": None,
        "new_api_calls": 0,
        "product_regression_tests_executed": 0,
        "done_item_count": 7,
        "open_item_count": 5,
        "selected_version_gate": "BLOCKED",
        "next_product_dependency": "VERIFY_EXECUTION_COST_BOUND_AND_DISPATCH_ONLY_THIS_REQUEST",
    }
    result["output"] = save("C156.R11_target_deepseek_request_preparation", result)
    print(json.dumps(result, ensure_ascii=False, indent=2))
    return result


V3_C156 = run_v3_c156()
````

### سل ۱۵۷ — ارسال کنترل‌شدهٔ یک درخواست DeepSeek

بدنهٔ پین‌شدهٔ C156، مدل، سقف خروجی و مسیر اتصال کنترل می‌شوند.
ارسال فقط با نرخ جاری ثبت‌شده و برآورد زیر سقف مجاز انجام می‌شود.
رسید شروع پیش از ارسال و بدنهٔ پاسخ پس از دریافت ذخیره می‌شوند.
اجرای مجدد، پاسخ موجود را مصرف می‌کند و وضعیت نامعلوم را باز می‌گذارد.

دامنه یک سؤال Test با رتبه‌های ثابت است. پاسخ دریافت‌شده برای داوری
سناریو آماده می‌شود و رأی رگرسیون در این سل صادر نمی‌شود.

````python
# C157 — R11.3 — Cost-guarded single dispatch of the pinned DeepSeek request
def run_v3_c157():
    def require(ok, code):
        if not ok:
            raise ValueError(code)

    def encode(value):
        return json.dumps(
            value, ensure_ascii=False, sort_keys=True,
            separators=(",", ":"), allow_nan=False,
        ).encode("utf-8")

    def sha(raw):
        return hashlib.sha256(raw).hexdigest()

    reports = Path("reports/phase2/quality_improvement/NB22E.lab.v3")
    cwd = Path.cwd()
    root = next((p.resolve() for p in (cwd, *cwd.parents)
                 if (p / reports).is_dir()), None)
    require(root is not None, "C157_PROJECT_ROOT_NOT_FOUND")

    def project_path(relative):
        relative = Path(relative)
        path = (root / relative).resolve()
        require(not relative.is_absolute() and path.is_relative_to(root),
                "C157_PATH_OUTSIDE_PROJECT")
        return path

    def load(relative, digest):
        raw = project_path(relative).read_bytes()
        require(sha(raw) == digest, f"C157_HASH_CONFLICT:{relative}")
        return json.loads(raw)

    def write_new(path, raw):
        with path.open("xb") as handle:
            handle.write(raw)
            handle.flush()
            os.fsync(handle.fileno())
        require(path.read_bytes() == raw, "C157_WRITE_RELOAD_CONFLICT")

    c156_sha = "ec4cd70abcd81fdae2659620da30edcf108d940df1e8a1bbed78fb5d7224730d"
    prepared = load(
        reports / f"C156.R11_target_deepseek_request_preparation.{c156_sha}.json",
        c156_sha,
    )
    ref = prepared["request_receipt"]
    job = load(ref["path"], ref["file_bytes_sha256"])
    keys = ("candidate_id", "question_sha256", "regression_spec_id")
    identity = {key: prepared[key] for key in keys}
    request = job["request"]
    body = encode(request)
    request_sha = sha(body)

    require(
        prepared["status"] == "TARGET_DEEPSEEK_REQUEST_PREPARED_OFFLINE"
        and job["dispatch_performed"] is False
        and all(job[key] == value for key, value in identity.items())
        and job["rendered_context_sha256"] == prepared["rendered_context_sha256"]
        and request_sha == job["request_payload_sha256"] == prepared["request_payload_sha256"]
        and request_sha != prepared["prior_request_payload_sha256"]
        and len(body) == prepared["request_bytes"]
        and set(request) == {"model", "messages", "temperature", "max_tokens", "thinking"}
        and request["model"] == "DeepSeek-V4-Flash"
        and request["temperature"] == 0
        and request["max_tokens"] == 600
        and request["thinking"] == {"type": "disabled"}
        and [message["role"] for message in request["messages"]] == ["system", "user"],
        "C157_PINNED_REQUEST_CONFLICT",
    )

    budget = 100_000
    input_estimate = len(body) + 1024
    output_cap = request["max_tokens"]
    historical_estimate = round(
        (input_estimate * 43_680 + output_cap * 87_360) / 1_000_000, 2
    )
    result = {
        "cell": "C157",
        "roadmap_action": "R11",
        "roadmap_subtask": "R11.3",
        "scope": "ONE_TEST_TARGET_FIXED_RANKS_DEEPSEEK",
        **identity,
        "input_file_hashes": {"C156": c156_sha, "C156_request": ref["file_bytes_sha256"]},
        "request_payload_sha256": request_sha,
        "configured_model": request["model"],
        "estimated_input_token_allowance": input_estimate,
        "requested_max_output_tokens": output_cap,
        "budget_toman": budget,
        "historical_reference_estimate_toman": historical_estimate,
        "current_rate_verified": False,
        "estimated_cost_toman": None,
        "new_api_calls_attempted": 0,
        "responses_received_and_checked": 0,
        "product_regression_tests_executed": 0,
        "done_item_count": 7,
        "open_item_count": 5,
        "selected_version_gate": "BLOCKED",
    }

    stage = "CHECK_EXISTING_ATTEMPT"
    try:
        # Key by the exact request so a gateway change cannot trigger a resend.
        receipt_dir = project_path(reports / "C157.R11_target_deepseek_receipts")
        receipt_dir.mkdir(exist_ok=True)
        started_path = receipt_dir / f"{request_sha}.started.json"
        response_path = receipt_dir / f"{request_sha}.response.json"

        if not response_path.exists():
            require(not started_path.exists(), "C157_OUTCOME_UNKNOWN_NO_RESEND")

            stage = "BIND_LOCAL_GATEWAY_AND_CURRENT_RATES"
            parser = globals().get("dotenv_values")
            url_module = globals().get("urllib")
            require(callable(parser) and hasattr(url_module, "parse")
                    and hasattr(url_module, "request") and hasattr(url_module, "error"),
                    "C157_C042_DOTENV_AND_URL_BINDINGS_REQUIRED")
            config = parser(root / ".env", interpolate=False)

            def setting(name):
                return str(os.environ.get(name) or config.get(name) or "").strip()

            input_rate = setting("R11_DEEPSEEK_INPUT_TOMAN_PER_MILLION")
            output_rate = setting("R11_DEEPSEEK_OUTPUT_TOMAN_PER_MILLION")
            rate_source = setting("R11_DEEPSEEK_RATE_SOURCE")
            require(
                input_rate.isascii() and input_rate.isdecimal()
                and output_rate.isascii() and output_rate.isdecimal()
                and int(input_rate) > 0 and int(output_rate) > 0 and bool(rate_source),
                "C157_CURRENT_DEEPSEEK_RATES_OR_SOURCE_MISSING",
            )
            input_rate, output_rate = int(input_rate), int(output_rate)
            estimate_units = input_estimate * input_rate + output_cap * output_rate
            result.update(
                rate_basis="OPERATOR_CONFIGURED_CURRENT_RATES",
                input_toman_per_million=input_rate,
                output_toman_per_million=output_rate,
                estimated_cost_toman=estimate_units / 1_000_000,
            )
            require(estimate_units < budget * 1_000_000, "C157_ESTIMATE_REACHES_BUDGET")

            base = str(config.get("ARVAN_GATEWAY_URL") or "").strip()
            key = str(config.get("ARVAN_GATEWAY_API_KEY") or "").strip()
            url = url_module.parse.urlsplit(base)
            require(
                url.scheme == "https" and bool(url.hostname)
                and url.username is None and url.password is None
                and not url.query and not url.fragment and bool(key),
                "C157_INVALID_GATEWAY_CONFIGURATION",
            )
            route = url.path.rstrip("/")
            if not route:
                route = "/v1/chat/completions"
            elif not route.endswith("/chat/completions"):
                route += "/chat/completions"
            endpoint = url._replace(path=route).geturl()

            class NoRedirect(url_module.request.HTTPRedirectHandler):
                def redirect_request(self, req, fp, code, msg, headers, newurl):
                    return None

            opener = url_module.request.build_opener(NoRedirect())
            outgoing = url_module.request.Request(
                endpoint, data=body, method="POST",
                headers={"Authorization": f"Bearer {key}",
                         "Content-Type": "application/json"},
            )

            stage = "PERSIST_INTENT_AND_SEND_ONCE"
            started = {
                **identity,
                "request_payload_sha256": request_sha,
                "request_receipt_sha256": ref["file_bytes_sha256"],
                "endpoint_sha256": sha(endpoint.encode("utf-8")),
                "input_toman_per_million": input_rate,
                "output_toman_per_million": output_rate,
                "rate_source": rate_source,
                "estimated_cost_toman": result["estimated_cost_toman"],
            }
            try:
                write_new(started_path, encode(started))
            except FileExistsError:
                raise ValueError("C157_EXISTING_ATTEMPT_NO_RESEND") from None

            result["new_api_calls_attempted"] = 1
            try:
                with opener.open(outgoing, timeout=90) as response:
                    http_status = response.status
                    response_text = response.read().decode("utf-8")
            except url_module.error.HTTPError as error:
                with error:
                    http_status = error.code
                    response_text = error.read().decode("utf-8")
            except (OSError, ValueError):
                raise ValueError("C157_OUTCOME_UNKNOWN_AFTER_DISPATCH_NO_RESEND") from None

            write_new(response_path, encode({
                **identity,
                "request_payload_sha256": request_sha,
                "started_receipt_sha256": sha(started_path.read_bytes()),
                "http_status": http_status,
                "response_body_sha256": sha(response_text.encode("utf-8")),
                "response_body": response_text,
            }))

        stage = "VERIFY_PERSISTED_RESPONSE"
        started_raw = started_path.read_bytes()
        started = json.loads(started_raw)
        response_raw = response_path.read_bytes()
        response = json.loads(response_raw)
        require(
            all(started[key] == response[key] == value for key, value in identity.items())
            and started["request_payload_sha256"] == response["request_payload_sha256"] == request_sha
            and started["request_receipt_sha256"] == ref["file_bytes_sha256"]
            and response["started_receipt_sha256"] == sha(started_raw)
            and sha(response["response_body"].encode("utf-8")) == response["response_body_sha256"],
            "C157_PERSISTED_RESPONSE_BINDING_CONFLICT",
        )
        result.update(
            http_status=response["http_status"],
            estimated_cost_toman=started["estimated_cost_toman"],
            response_receipt={
                "path": response_path.relative_to(root).as_posix(),
                "file_bytes_sha256": sha(response_raw),
            },
        )
        require(response["http_status"] == 200, "C157_NON200_RESPONSE_PRESERVED_NO_RESEND")
        envelope = json.loads(response["response_body"])
        actual_model = envelope["model"].strip().rsplit("/", 1)[-1].lower()
        choice = envelope["choices"][0]
        answer = choice["message"]["content"]
        require(
            actual_model in {"deepseek-v4-flash", "deepseek-v4-flash-0731"}
            and isinstance(answer, str) and bool(answer.strip()),
            "C157_RESPONSE_MODEL_OR_CONTENT_CONFLICT",
        )
        result.update(
            status="TARGET_DEEPSEEK_RESPONSE_BOUND_FOR_REVIEW",
            blocker=None,
            responses_received_and_checked=1,
            response_model=actual_model,
            finish_reason=choice.get("finish_reason"),
            response_text_sha256=sha(answer.encode("utf-8")),
            usage_as_reported=envelope.get("usage"),
            next_product_dependency="ASSESS_ANSWER_SUPPORTED_AGAINST_BOUND_EVIDENCE",
        )
    except (OSError, ValueError, KeyError, TypeError, IndexError) as error:
        code = str(error)
        result.update(
            status="TARGET_DEEPSEEK_EXECUTION_BLOCKED",
            blocker={"stage": stage, "code": code if code.startswith("C157_")
                     else type(error).__name__},
        )

    raw = encode(result)
    digest = sha(raw)
    relative = reports / f"C157.R11_target_deepseek_execution.{digest}.json"
    target = project_path(relative)
    try:
        write_new(target, raw)
    except FileExistsError:
        require(target.read_bytes() == raw, "C157_REPORT_RELOAD_CONFLICT")
    result["output"] = {"path": relative.as_posix(), "file_bytes_sha256": digest,
                        "reload_validation": "PASS"}
    print(json.dumps(result, ensure_ascii=False, indent=2))
    return result


V3_C157 = run_v3_c157()
````

````python
# C168 — R11.3 — Bind existing evidence for the seven diagnostic residuals
def run_v3_c168():
    def require(condition, code):
        if not condition:
            raise ValueError(code)

    def sha(raw):
        return hashlib.sha256(raw).hexdigest()

    reports = Path("reports/phase2/quality_improvement/NB22E.lab.v3")
    cwd = Path.cwd()
    root = next(
        (p.resolve() for p in (cwd, *cwd.parents) if (p / reports).is_dir()),
        None,
    )
    require(root is not None, "C168_PROJECT_ROOT_NOT_FOUND")

    pins = {
        "C072": (
            "C072.R11_diagnostic_evidence_dispositions",
            "53835a42f7d8071d31499b95391f8f4d20a466cd86130b51f244be2698e0d370",
        ),
        "C073": (
            "C073.R11_conditional_response_comparisons",
            "34f7316e669852c83af56a3a33d5a7eb2a6abdce197c419193b933ba44e4ad2c",
        ),
        "C080": (
            "C080.R11_diagnostic_residuals_and_execution_requirements",
            "15dba7dddba32ec40bd63547637735794932a72c02a970df4e45b1f49678857b",
        ),
        "C167": (
            "C167.R11_target_conditional_content_checks",
            "f7dcda3a87276f25e2fe7027c1a62f38eb1c6777bb407a74fe362f545536026b",
        ),
    }

    def path_for(relative):
        path = (root / relative).resolve()
        require(path.is_relative_to(root), "C168_PATH_OUTSIDE_PROJECT")
        return path

    def load(label):
        stem, digest = pins[label]
        raw = path_for(reports / f"{stem}.{digest}.json").read_bytes()
        require(sha(raw) == digest, f"C168_HASH_CONFLICT:{label}")
        return json.loads(raw)

    docs = {label: load(label) for label in pins}
    base = docs["C080"]
    savings = docs["C167"]
    pending = base["pending_regression_spec_ids"]
    require(
        base["status"] == "DIAGNOSTIC_RESIDUALS_AND_PRODUCT_EXECUTION_REQUIREMENTS_RECORDED"
        and len(pending) == len(set(pending))
            == base["batch_scenarios_pending_final_evidence_sufficiency"] == 7
        and set(docs["C072"]["pending_regression_spec_ids"]) == set(pending)
        and set(docs["C073"]["pending_regression_spec_ids"]) == set(pending)
        and docs["C073"]["input_file_hashes"]["C072"] == pins["C072"][1]
        and savings["status"] == "TWO_CONDITIONAL_CONTENT_CHECKS_BOUND_TO_TARGET"
        and savings["regression_spec_id"] not in pending
        and savings["component_comparison_status"] == "NOT_COMPARABLE_NOT_PASS",
        "C168_RESIDUAL_OR_SOURCE_BINDING_CONFLICT",
    )

    def walk(value, pointer=""):
        if isinstance(value, dict):
            yield pointer, value
            for key, child in value.items():
                token = key.replace("~", "~0").replace("/", "~1")
                yield from walk(child, f"{pointer}/{token}")
        elif isinstance(value, list):
            for index, child in enumerate(value):
                yield from walk(child, f"{pointer}/{index}")

    def index_records(document, wanted):
        indexed = {spec_id: [] for spec_id in wanted}
        for pointer, row in walk(document):
            ids = {
                row[key]
                for key in ("regression_spec_id", "spec_id", "regression_id")
                if isinstance(row.get(key), str)
            }
            pointer_key = pointer.rsplit("/", 1)[-1]
            if pointer_key in wanted:
                ids.add(pointer_key)
            require(
                len(ids) <= 1 or not ids.intersection(wanted),
                f"C168_CONFLICTING_RECORD_IDENTITY:{pointer}",
            )
            for spec_id in ids.intersection(wanted):
                indexed[spec_id].append({"json_pointer": pointer, "record": row})
        return indexed

    wanted = set(pending)
    indexes = {
        label: index_records(docs[label], wanted) for label in ("C072", "C073")
    }
    rows = []
    for spec_id in pending:
        evidence = {
            label: records[spec_id] for label, records in indexes.items()
            if records[spec_id]
        }
        rows.append({
            "regression_spec_id": spec_id,
            "state": "PENDING_FINAL_EVIDENCE_SUFFICIENCY",
            "bound_existing_records": evidence,
        })

    missing = [
        row["regression_spec_id"] for row in rows if not row["bound_existing_records"]
    ]
    result = {
        "cell": "C168",
        "roadmap_action": "R11",
        "roadmap_subtask": "R11.3",
        "status": (
            "DIAGNOSTIC_RESIDUAL_EVIDENCE_BOUND"
            if not missing else "DIAGNOSTIC_RESIDUAL_EVIDENCE_PARTIALLY_BOUND"
        ),
        "input_file_hashes": {label: digest for label, (_, digest) in pins.items()},
        "pending_answer_evidence_sufficiency_count": len(rows),
        "residual_rows_with_explicit_evidence_binding": len(rows) - len(missing),
        "residual_spec_ids_without_explicit_record": missing,
        "answer_evidence_residuals": rows,
        "corporate_profit_role_distinction_open": base["corporate_profit_role_distinction_open"],
        "savings_scenario": {
            "regression_spec_id": savings["regression_spec_id"],
            "conditional_checks_preserved": len(savings["conditional_content_checks"]),
            "component_comparison_status": savings["component_comparison_status"],
        },
        "new_semantic_judgments": 0,
        "product_regression_tests_executed": 0,
        "new_api_calls": 0,
        "done_item_count": savings["done_item_count"],
        "open_item_count": savings["open_item_count"],
        "newly_completed_items_from_previous_open_set": [],
        "historical_gate": savings["historical_gate"],
        "selected_version_gate": savings["selected_version_gate"],
        "next_product_dependency": "ASSESS_SEVEN_BOUND_RESIDUALS_AND_KEEP_UNSUPPORTED_DISPOSITIONS_OPEN",
    }

    raw = json.dumps(
        result, ensure_ascii=False, sort_keys=True,
        separators=(",", ":"), allow_nan=False,
    ).encode("utf-8")
    digest = sha(raw)
    relative = reports / f"C168.R11_existing_diagnostic_residual_evidence.{digest}.json"
    path = path_for(relative)
    try:
        with path.open("xb") as handle:
            handle.write(raw)
            handle.flush()
            os.fsync(handle.fileno())
    except FileExistsError:
        pass
    require(path.read_bytes() == raw, "C168_REPORT_RELOAD_CONFLICT")
    result["output"] = {
        "path": relative.as_posix(),
        "file_bytes_sha256": digest,
        "reload_validation": "PASS",
    }
    print(json.dumps(result, ensure_ascii=False, indent=2))
    return result


V3_C168 = run_v3_c168()
````

### سل ۱۶۹ — تعیین وضعیت تشخیصی هفت پروندهٔ باز

شرکت جدا باز می‌مانند. هیچ API یا بازداوری عمومی انجام نمی‌شود.

````python
# C169 — R11.3 — Record scoped dispositions for the seven existing residuals
def run_v3_c169():
    def require(ok, code):
        if not ok:
            raise ValueError(code)

    def sha(raw):
        return hashlib.sha256(raw).hexdigest()

    reports = Path("reports/phase2/quality_improvement/NB22E.lab.v3")
    cwd = Path.cwd()
    root = next(
        (p.resolve() for p in (cwd, *cwd.parents) if (p / reports).is_dir()),
        None,
    )
    require(root is not None, "C169_PROJECT_ROOT_NOT_FOUND")

    c168_sha = "c8789c128b7a966c9b5b33346f338e0922744b7c9f6bdce4c05a993da5a34e7e"
    source = reports / f"C168.R11_existing_diagnostic_residual_evidence.{c168_sha}.json"
    raw = (root / source).read_bytes()
    require(sha(raw) == c168_sha, "C169_C168_HASH_CONFLICT")
    prior = json.loads(raw)

    residuals = prior["answer_evidence_residuals"]
    require(
        prior["status"] == "DIAGNOSTIC_RESIDUAL_EVIDENCE_BOUND"
        and prior["pending_answer_evidence_sufficiency_count"]
            == prior["residual_rows_with_explicit_evidence_binding"]
            == len(residuals) == 7
        and len({row["regression_spec_id"] for row in residuals}) == 7,
        "C169_RESIDUAL_POPULATION_CONFLICT",
    )

    requirements = {
        "CONDITIONAL_ANSWER_SUPPORTED": (
            "RESOLVE_INTENDED_REFERENT_BEFORE_UNCONDITIONAL_VERDICT",
            "روشن شود برداشت دارای شاهد همان مرجع سؤال است یا پاسخ باید مشروط بماند.",
        ),
        "ANSWER_COMPONENTS_SUPPORTED": (
            "ASSESS_EVIDENCE_FOR_SPECIFIC_RECOMMENDATION_IF_REQUIRED",
            "اگر انتظار آزمون توصیهٔ مشخص است، کفایت شاهد برای همان توصیه بررسی شود.",
        ),
        "CHECKED_PASSAGES_INCONCLUSIVE": (
            "ASSESS_REMAINING_SUPPLIED_EVIDENCE_BEFORE_SUFFICIENCY_VERDICT",
            "پیش از حکم دربارهٔ کفایت، شواهد ارسالی خارج از قطعه‌های بررسی‌شده لحاظ شوند.",
        ),
    }
    counts = dict.fromkeys(requirements, 0)
    rows = []

    for residual in residuals:
        spec_id = residual["regression_spec_id"]
        records = residual["bound_existing_records"]
        require(
            len(records.get("C072", [])) == len(records.get("C073", [])) == 1,
            f"C169_EXPECTED_ONE_RECORD_PER_SOURCE:{spec_id}",
        )
        before, after = (records[label][0] for label in ("C072", "C073"))
        old, current = before["record"], after["record"]
        disposition = current["diagnostic_batch_disposition"]
        evidence = current["diagnostic_evidence_sufficiency_review"]
        require(
            old["candidate_id"] == current["candidate_id"]
            and old["diagnostic_batch_disposition"] == disposition
            and disposition["final_scenario_verdict"] is None
            and evidence["final_scenario_verdict"] is None
            and evidence["full_supplied_evidence_sufficiency_assessed"] is False,
            f"C169_SCOPE_OR_IDENTITY_CONFLICT:{spec_id}",
        )

        status = disposition["status"]
        require(status in requirements, f"C169_UNEXPECTED_DISPOSITION:{spec_id}")
        counts[status] += 1
        gap, gap_fa = requirements[status]
        comparison = current.get("diagnostic_conditional_response_comparison")
        if status == "CONDITIONAL_ANSWER_SUPPORTED":
            require(
                isinstance(comparison, dict)
                and comparison["intended_question_referent_resolved"] is False
                and comparison["final_scenario_verdict"] is None
                and comparison["generation_input_sha256"] == evidence["generation_input_sha256"]
                and comparison["response_content_sha256"] == evidence["response_content_sha256"],
                f"C169_CONDITIONAL_COMPARISON_CONFLICT:{spec_id}",
            )

        rows.append({
            "regression_spec_id": spec_id,
            "candidate_id": current["candidate_id"],
            "diagnostic_disposition": status,
            "reason_fa": disposition["reason_fa"],
            "bounded_answer_fa": evidence.get("bounded_answer_fa"),
            "recorded_response_comparison": (
                {"status": comparison["status"], "reason_fa": comparison["reason_fa"]}
                if comparison is not None else None
            ),
            "remaining_evidence_gap": gap,
            "remaining_evidence_gap_fa": gap_fa,
            "source_pointers": {"C072": before["json_pointer"], "C073": after["json_pointer"]},
            "final_scenario_verdict": None,
        })

    require(
        counts == {
            "CONDITIONAL_ANSWER_SUPPORTED": 3,
            "ANSWER_COMPONENTS_SUPPORTED": 1,
            "CHECKED_PASSAGES_INCONCLUSIVE": 3,
        },
        "C169_DISPOSITION_COUNTS_CONFLICT",
    )

    result = {
        "cell": "C169",
        "roadmap_action": "R11",
        "roadmap_subtask": "R11.3",
        "status": "SEVEN_SCOPED_DIAGNOSTIC_DISPOSITIONS_RECORDED",
        "input_file_hashes": {"C168": c168_sha, **prior["input_file_hashes"]},
        "diagnostic_disposition_counts": counts,
        "answer_evidence_residuals": rows,
        "final_evidence_sufficiency_open_count": len(rows),
        "corporate_profit_role_distinction_open": prior["corporate_profit_role_distinction_open"],
        "savings_scenario": prior["savings_scenario"],
        "new_semantic_judgments": 0,
        "new_api_calls": 0,
        "product_regression_tests_executed": 0,
        "done_item_count": prior["done_item_count"],
        "open_item_count": prior["open_item_count"],
        "newly_completed_items_from_previous_open_set": [],
        "historical_gate": prior["historical_gate"],
        "selected_version_gate": prior["selected_version_gate"],
        "next_product_dependency": "INSPECT_FULL_DISPATCHED_EVIDENCE_FOR_THREE_INCONCLUSIVE_CASES",
    }
    payload = json.dumps(
        result, ensure_ascii=False, sort_keys=True,
        separators=(",", ":"), allow_nan=False,
    ).encode("utf-8")
    digest = sha(payload)
    relative = reports / f"C169.R11_scoped_diagnostic_residual_dispositions.{digest}.json"
    target = (root / relative).resolve()
    require(target.is_relative_to(root), "C169_OUTPUT_PATH_OUTSIDE_PROJECT")
    try:
        with target.open("xb") as handle:
            handle.write(payload)
            handle.flush()
            os.fsync(handle.fileno())
    except FileExistsError:
        pass
    require(target.read_bytes() == payload, "C169_REPORT_RELOAD_CONFLICT")
    result["output"] = {
        "path": relative.as_posix(),
        "file_bytes_sha256": digest,
        "reload_validation": "PASS",
    }
    print(json.dumps(result, ensure_ascii=False, indent=2))
    return result


V3_C169 = run_v3_c169()
````

### سل ۱۷۰ — خواندن شواهد کامل سه پروندهٔ بی‌نتیجه

سه شناسه از C169 انتخاب می‌شوند. برای هر پرونده، متن کامل درخواست
از طرح پین‌شدهٔ C040 و پاسخ از C043 و رسید C042 خوانده می‌شود.
اتصال درخواست به رسید قبلی با همان قواعد C154 بررسی می‌شود.

این سل متن لازم برای داوری را آماده می‌کند. رأی کفایت شواهد در این
مرحله باز می‌ماند و هیچ درخواست تازه‌ای ارسال نمی‌شود.

</div>

````python
# C170 — R11.3 — Read the complete dispatched inputs for three residuals
def run_v3_c170():
    def require(ok, code):
        if not ok:
            raise ValueError(code)

    def sha(raw):
        return hashlib.sha256(raw).hexdigest()

    def encode(value):
        return json.dumps(
            value, ensure_ascii=False, sort_keys=True,
            separators=(",", ":"), allow_nan=False,
        ).encode("utf-8")

    reports = Path("reports/phase2/quality_improvement/NB22E.lab.v3")
    cwd = Path.cwd()
    root = next(
        (p.resolve() for p in (cwd, *cwd.parents) if (p / reports).is_dir()),
        None,
    )
    require(root is not None, "C170_PROJECT_ROOT_NOT_FOUND")

    def path_for(relative):
        relative = Path(relative)
        path = (root / relative).resolve()
        require(
            not relative.is_absolute() and path.is_relative_to(root),
            "C170_PATH_OUTSIDE_PROJECT",
        )
        return path

    def load(relative, digest):
        raw = path_for(relative).read_bytes()
        require(sha(raw) == digest, f"C170_HASH_CONFLICT:{relative}")
        return json.loads(raw)

    pins = {
        "C169": (
            "C169.R11_scoped_diagnostic_residual_dispositions."
            "067e624832446e564b64f32596801de3cbd5f52156c572a553566ab2c59418fe.json",
            "067e624832446e564b64f32596801de3cbd5f52156c572a553566ab2c59418fe",
        ),
        "C040_plan": (
            "C040.R11_generation_plan."
            "608d2551794d2028f973e23a89e96079b0f8b2fead857f43c16079379f29b3bd.json",
            "cda83957141dcfbf85e2073b9af21e65d8406da2e328da6c2b5f25a713579e85",
        ),
        "C042": (
            "C042.R11_deepseek_batch."
            "e8ff5073b9a0c595dda7a0cd7d49975518dac17c03c9130b9ae3c3cd35b60e8b.json",
            "e8ff5073b9a0c595dda7a0cd7d49975518dac17c03c9130b9ae3c3cd35b60e8b",
        ),
        "C043": (
            "C043.R11_scenario_review_inputs."
            "811af55fabd63d695bd2767c762055c4997d81e4acd7c36f7bdde717421e6145.json",
            "811af55fabd63d695bd2767c762055c4997d81e4acd7c36f7bdde717421e6145",
        ),
    }
    docs = {key: load(reports / name, digest)
            for key, (name, digest) in pins.items()}
    prior = docs["C169"]
    selected = [
        row for row in prior["answer_evidence_residuals"]
        if row["diagnostic_disposition"] == "CHECKED_PASSAGES_INCONCLUSIVE"
    ]
    require(
        prior["status"] == "SEVEN_SCOPED_DIAGNOSTIC_DISPOSITIONS_RECORDED"
        and len(selected) == len({row["candidate_id"] for row in selected}) == 3
        and len({row["regression_spec_id"] for row in selected}) == 3
        and all(row["final_scenario_verdict"] is None for row in selected),
        "C170_RESIDUAL_POPULATION_CONFLICT",
    )

    config = dotenv_values(root / ".env", interpolate=False)
    base = str(config.get("ARVAN_GATEWAY_URL") or "").strip()
    url = urllib.parse.urlsplit(base)
    require(
        url.scheme == "https" and bool(url.hostname)
        and url.username is None and url.password is None
        and not url.query and not url.fragment,
        "C170_GATEWAY_IDENTITY_UNAVAILABLE",
    )
    route = url.path.rstrip("/")
    if not route:
        route = "/v1/chat/completions"
    elif not route.endswith("/chat/completions"):
        route += "/chat/completions"
    endpoint_sha = sha(url._replace(path=route).geturl().encode("utf-8"))

    def one_row(rows, candidate, label):
        matches = [(i, row) for i, row in enumerate(rows)
                   if row["candidate_id"] == candidate]
        require(len(matches) == 1, f"C170_EXPECTED_ONE_ROW:{label}:{candidate}")
        return matches[0]

    cases = []
    gaps = []
    for target in selected:
        candidate = target["candidate_id"]
        spec_id = target["regression_spec_id"]
        try:
            job_index, job = one_row(docs["C040_plan"]["jobs"], candidate, "C040")
            batch_index, batch = one_row(docs["C042"]["receipts"], candidate, "C042")
            review_index, review = one_row(docs["C043"]["responses"], candidate, "C043")
            require(
                job["job_sha256"] == review["source_plan_job_sha256"]
                and job["generation_input_sha256"] == review["generation_input_sha256"],
                "C170_PLAN_JOB_BINDING_CONFLICT",
            )
            messages = job["request"]["messages"]
            require(
                [message["role"] for message in messages] == ["system", "user"]
                and all(isinstance(message["content"], str) and message["content"]
                        for message in messages),
                "C170_MESSAGE_STRUCTURE_CONFLICT",
            )
            payload = {
                "model": "DeepSeek-V4-Flash",
                "messages": messages,
                "temperature": 0,
                "max_tokens": 600,
                "thinking": {"type": "disabled"},
            }
            payload_sha = sha(encode(payload))
            request_id = sha(encode({
                "candidate_id": candidate,
                "plan_sha256": pins["C040_plan"][1],
                "endpoint_sha256": endpoint_sha,
                "request_sha256": payload_sha,
            }))
            require(request_id == review["request_id"],
                    "C170_HISTORICAL_REQUEST_ID_NOT_REPRODUCED")

            directory = reports / "C042.R11_deepseek_receipts"
            response_path = directory / f"{request_id}.response.json"
            require(
                batch["receipt_path"] == review["receipt_path"] == response_path.as_posix(),
                "C170_RESPONSE_PATH_CONFLICT",
            )
            receipt = load(response_path, review["receipt_file_bytes_sha256"])
            started = json.loads(path_for(
                directory / f"{request_id}.started.json"
            ).read_bytes())
            require(
                receipt["status"] == "HTTP_200_RECEIVED"
                and receipt["request_id"] == started["request_id"] == request_id
                and receipt["candidate_id"] == started["candidate_id"] == candidate,
                "C170_RESPONSE_RECEIPT_IDENTITY_CONFLICT",
            )
            response_body = receipt["response_body"]
            require(
                sha(response_body.encode("utf-8")) == receipt["response_bytes_sha256"]
                    == batch["response_bytes_sha256"] == review["response_bytes_sha256"],
                "C170_RESPONSE_BODY_HASH_CONFLICT",
            )
            envelope = json.loads(response_body)
            answer = envelope["choices"][0]["message"]["content"]
            require(
                isinstance(answer, str) and bool(answer.strip())
                and answer == review["response_content"]
                and envelope["model"].strip().rsplit("/", 1)[-1].lower()
                    == batch["response_model"],
                "C170_RESPONSE_CONTENT_OR_MODEL_CONFLICT",
            )
            cases.append({
                "regression_spec_id": spec_id,
                "candidate_id": candidate,
                "previous_diagnostic_reason_fa": target["reason_fa"],
                "previous_review_pointers": target["source_pointers"],
                "source_pointers": {
                    "C040_plan": f"/jobs/{job_index}",
                    "C042": f"/receipts/{batch_index}",
                    "C043": f"/responses/{review_index}",
                },
                "generation_input_sha256": job["generation_input_sha256"],
                "request_payload_sha256": payload_sha,
                "system_message_sha256": sha(messages[0]["content"].encode("utf-8")),
                "exact_dispatched_user_message": messages[1]["content"],
                "response": answer,
                "response_text_sha256": sha(answer.encode("utf-8")),
                "final_evidence_sufficiency_verdict": None,
            })
        except (OSError, ValueError, KeyError, TypeError, IndexError) as error:
            gaps.append({
                "regression_spec_id": spec_id,
                "candidate_id": candidate,
                "binding_error": str(error),
            })

    result = {
        "cell": "C170",
        "roadmap_action": "R11",
        "roadmap_subtask": "R11.3",
        "status": (
            "THREE_FULL_DISPATCHED_INPUTS_BOUND_FOR_REVIEW"
            if len(cases) == 3 else "FULL_DISPATCHED_INPUT_BINDING_INCOMPLETE"
        ),
        "input_file_hashes": {key: digest for key, (_, digest) in pins.items()},
        "target_count": 3,
        "bound_count": len(cases),
        "cases": cases,
        "binding_gaps": gaps,
        "new_semantic_judgments": 0,
        "new_api_calls": 0,
        "product_regression_tests_executed": 0,
        "done_item_count": prior["done_item_count"],
        "open_item_count": prior["open_item_count"],
        "historical_gate": prior["historical_gate"],
        "selected_version_gate": prior["selected_version_gate"],
        "next_product_dependency": (
            "REVIEW_BOUND_FULL_EVIDENCE_FOR_THE_THREE_RESIDUALS"
            if len(cases) == 3 else "RESOLVE_RECORDED_REQUEST_OR_RESPONSE_BINDING_GAPS"
        ),
    }
    raw = encode(result)
    digest = sha(raw)
    relative = reports / f"C170.R11_three_residual_full_dispatched_inputs.{digest}.json"
    path = path_for(relative)
    try:
        with path.open("xb") as handle:
            handle.write(raw)
            handle.flush()
            os.fsync(handle.fileno())
    except FileExistsError:
        pass
    require(path.read_bytes() == raw, "C170_REPORT_RELOAD_CONFLICT")
    result["output"] = {
        "path": relative.as_posix(),
        "file_bytes_sha256": digest,
        "reload_validation": "PASS",
    }
    print(json.dumps(result, ensure_ascii=False, indent=2))
    return result


V3_C170 = run_v3_c170()
````

### سل ۱۷۱ — ثبت داوری سه پاسخ با شواهد کامل ارسالی

````python
# C172 — R11.3 — Construct the controlled incumbent arm and check exact reuse
def run_v3_c172():
    def require(ok, code):
        if not ok:
            raise ValueError(code)

    def encode(value):
        return json.dumps(
            value, ensure_ascii=False, sort_keys=True,
            separators=(",", ":"), allow_nan=False,
        ).encode("utf-8")

    def sha(raw):
        return hashlib.sha256(raw).hexdigest()

    def text_sha(text):
        return sha(text.encode("utf-8"))

    reports = Path("reports/phase2/quality_improvement/NB22E.lab.v3")
    cwd = Path.cwd()
    root = next(
        (p.resolve() for p in (cwd, *cwd.parents) if (p / reports).is_dir()),
        None,
    )
    require(root is not None, "C172_PROJECT_ROOT_NOT_FOUND")

    def path_for(relative):
        relative = Path(relative)
        path = (root / relative).resolve()
        require(not relative.is_absolute() and path.is_relative_to(root),
                "C172_PATH_OUTSIDE_PROJECT")
        return path

    def read_bound(relative, digest):
        raw = path_for(relative).read_bytes()
        require(sha(raw) == digest, f"C172_HASH_CONFLICT:{relative}")
        return raw

    def load(ref):
        return json.loads(read_bound(ref["path"], ref["file_bytes_sha256"]))

    def save(stem, value):
        raw = encode(value)
        digest = sha(raw)
        relative = reports / f"{stem}.{digest}.json"
        path = path_for(relative)
        try:
            with path.open("xb") as handle:
                handle.write(raw)
                handle.flush()
                os.fsync(handle.fileno())
        except FileExistsError:
            pass
        require(path.read_bytes() == raw, "C172_REPORT_RELOAD_CONFLICT")
        return {"path": relative.as_posix(), "file_bytes_sha256": digest,
                "reload_validation": "PASS"}

    pins = {
        "C146": (
            "C146.R11_target_retrieval_input_binding",
            "226aca0c92a8bd7f23f06b6714c7d427f65b46799a2b7fe6e9456cc7710eeb2e",
        ),
        "C151": (
            "C151.R11_target_generation_configuration_binding",
            "18e5feafd6c8dafceb754346bd0d22afadda5e42d10f33d80fc81459df903045",
        ),
        "C154": (
            "C154.R11_prior_deepseek_request_response_binding",
            "4c46b189e0649208455f099e1274c793a8fa333f57ccae1926785e3c238358ac",
        ),
        "C156": (
            "C156.R11_target_deepseek_request_preparation",
            "ec4cd70abcd81fdae2659620da30edcf108d940df1e8a1bbed78fb5d7224730d",
        ),
        "C164": (
            "C164.R11_target_computed_rank_replay_disposition",
            "e3d7e6f5538c884888843b96b705e6821713f7e4ae1e5af11d3d0a6a72332abd",
        ),
        "C166": (
            "C166.R11_target_comparative_plan_prerequisite",
            "b595338bdd7cd0e4201145c2305e3d6f671caf7efb521e0d20f1777ebbcac154",
        ),
        "C167": (
            "C167.R11_target_conditional_content_checks",
            "f7dcda3a87276f25e2fe7027c1a62f38eb1c6777bb407a74fe362f545536026b",
        ),
    }
    hashes = {cell: digest for cell, (_, digest) in pins.items()}
    docs = {
        cell: json.loads(read_bound(reports / f"{stem}.{digest}.json", digest))
        for cell, (stem, digest) in pins.items()
    }
    ranks, context_report, historical, prepared, replay, formal, criteria = (
        docs[cell] for cell in ("C146", "C151", "C154", "C156", "C164", "C166", "C167")
    )

    keys = ("candidate_id", "question_sha256", "regression_spec_id")
    identity = {key: replay[key] for key in keys}
    require(
        all(document[key] == value
            for document in docs.values() for key, value in identity.items())
        and ranks["status"] == "TARGET_HISTORICAL_HYBRID100_AND_POOL50_BOUND"
        and ranks["target_split"] == "test"
        and context_report["exact_context_verified"] is True
        and historical["prior_request_id_reproduced"] is True
        and historical["prior_response_bytes_verified"] is True
        and replay["target_rank_computation_inside_product_call_verified"] is True
        and replay["saved_response_replay_inside_same_product_call_verified"] is True
        and formal["input_file_hashes"]["C164"] == hashes["C164"]
        and criteria["input_file_hashes"]["C166"] == hashes["C166"]
        and formal["component_comparison_status"] == "NOT_COMPARABLE_NOT_PASS",
        "C172_PRIOR_EVIDENCE_CONFLICT",
    )

    selected = load(prepared["request_receipt"])
    context_record = load(context_report["prepared_input"])
    current_request = selected["request"]
    current_context = context_record["context"]
    question = context_record["question"]
    require(
        all(document[key] == value
            for document in (selected, context_record) for key, value in identity.items())
        and text_sha(question) == identity["question_sha256"]
        and text_sha(current_context) == selected["rendered_context_sha256"]
            == context_record["rendered_context_sha256"]
            == criteria["reviewed_context_sha256"]
        and sha(encode(current_request)) == selected["request_payload_sha256"]
            == prepared["request_payload_sha256"]
        and current_request["model"] == "DeepSeek-V4-Flash"
        and current_request["temperature"] == 0
        and current_request["max_tokens"] == 600
        and current_request["thinking"] == {"type": "disabled"},
        "C172_SELECTED_REQUEST_OR_CONTEXT_CONFLICT",
    )

    pool = ranks["pool50_order"]
    baseline_ids = pool[:10]
    selected_ids = selected["ordered_chunk_ids"]
    require(
        len(pool) == len(set(pool)) == 50
        and len(selected_ids) == len(set(selected_ids)) == 10
        and len(baseline_ids) == len(set(baseline_ids)) == 10,
        "C172_ARM_DEPTH_OR_ID_CONFLICT",
    )

    source_pins = context_report["input_file_hashes"]
    corpus_path = "data/phase2/chunks/nb11_benchmark_v1/baseline_phase1_char_safe_v1.jsonl"
    wanted = set(baseline_ids) | set(selected_ids)
    corpus = {}
    for line in read_bound(corpus_path, source_pins["chunks"]).splitlines():
        if not line.strip():
            continue
        row = json.loads(line)
        chunk_id = row["chunk_id"]
        if chunk_id in wanted:
            require(chunk_id not in corpus, "C172_DUPLICATE_CHUNK")
            corpus[chunk_id] = row
    require(set(corpus) == wanted, "C172_REQUIRED_CHUNK_MISSING")

    pipeline_path = "src/pipeline/rag_pipeline.py"
    tree = ast.parse(read_bound(pipeline_path, source_pins["pipeline_source"]),
                     filename=pipeline_path)
    pipeline_classes = [
        node for node in tree.body
        if isinstance(node, ast.ClassDef) and node.name == "RAGPipeline"
    ]
    require(len(pipeline_classes) == 1, "C172_PIPELINE_CLASS_CONFLICT")
    builders = [
        node for node in pipeline_classes[0].body
        if isinstance(node, ast.FunctionDef) and node.name == "build_context"
    ]
    require(len(builders) == 1 and not builders[0].decorator_list,
            "C172_CONTEXT_BUILDER_CONFLICT")
    namespace = {}
    exec(compile(
        ast.Module(body=builders, type_ignores=[]), pipeline_path, "exec",
        dont_inherit=True,
    ), namespace)

    def render(ids):
        chunks = [
            {"chunk_id": cid, "text": corpus[cid]["content"],
             "title": corpus[cid].get("title"), "url": corpus[cid].get("url")}
            for cid in ids
        ]
        return namespace["build_context"](None, chunks)

    require(render(selected_ids) == current_context,
            "C172_CONTEXT_BUILDER_DOES_NOT_REPRODUCE_SELECTED_ARM")
    baseline_context = render(baseline_ids)
    messages = current_request["messages"]
    require(
        [message["role"] for message in messages] == ["system", "user"]
        and bool(current_context)
        and messages[1]["content"].count(current_context) == 1,
        "C172_USER_CONTEXT_BOUNDARY_CONFLICT",
    )
    prefix, _, suffix = messages[1]["content"].partition(current_context)
    require(question in prefix + suffix, "C172_QUESTION_NOT_IN_FIXED_WRAPPER")

    baseline_request = {
        **current_request,
        "messages": [
            messages[0],
            {**messages[1], "content": prefix + baseline_context + suffix},
        ],
    }
    baseline_request_sha = sha(encode(baseline_request))
    historical_match = baseline_request_sha == historical["prior_request_payload_sha256"]
    same_as_selected = baseline_request_sha == prepared["request_payload_sha256"]

    if same_as_selected:
        next_step = "RECORD_NO_GENERATOR_INPUT_CONTRAST_FOR_THIS_TARGET"
    elif historical_match:
        next_step = "REVIEW_EXISTING_BASELINE_AND_SELECTED_ANSWERS_UNDER_SAME_CRITERIA"
    else:
        next_step = "ONE_BASELINE_RESPONSE_NEEDED_FOR_CONTROLLED_CASE_COMPARISON"

    baseline_ref = save("C172.R11_savings_controlled_baseline_request", {
        **identity,
        "scope": "ONE_TARGET_FIXED_RANKS_CONTROLLED_RETRIEVAL_COMPARISON",
        "baseline_arm": "HISTORICAL_RERANKER_TOP10",
        "pool50_source": ranks["source_rows"]["reranker"],
        "ordered_chunk_ids": baseline_ids,
        "ordered_chunk_text_sha256": [
            text_sha(corpus[cid]["content"]) for cid in baseline_ids
        ],
        "rendered_context_sha256": text_sha(baseline_context),
        "request_payload_sha256": baseline_request_sha,
        "request": baseline_request,
        "dispatch_performed": False,
    })

    result = {
        "cell": "C172",
        "roadmap_action": "R11",
        "roadmap_subtask": "R11.3",
        "status": "CONTROLLED_BASELINE_REQUEST_BUILT_AND_HISTORICAL_REUSE_CHECKED",
        **identity,
        "input_file_hashes": {
            **hashes,
            "C151_prepared_input": context_report["prepared_input"]["file_bytes_sha256"],
            "C156_request": prepared["request_receipt"]["file_bytes_sha256"],
            "pipeline_source": source_pins["pipeline_source"],
            "chunks": source_pins["chunks"],
        },
        "comparison_scope": "ONE_CASE_FIXED_INPUT_RETRIEVAL_ARM_COMPARISON",
        "variable": "ORDERED_TOP10_EVIDENCE_FROM_RERANKER_VS_CHAR_RRF",
        "fixed": [
            "QUESTION", "SYSTEM_MESSAGE", "USER_MESSAGE_WRAPPER", "CONTEXT_BUILDER",
            "TOP_K", "REQUESTED_MODEL", "GENERATION_PARAMETERS", "REVIEW_CRITERIA",
        ],
        "case_criteria": {
            "expected_behavior": criteria["expected_behavior"],
            "core_pass_fa": "هستهٔ پاسخ مستقیماً به سؤال پاسخ دهد و در شواهد همان بازو پشتوانه داشته باشد.",
            "detail_checks_apply_only_if_claim_is_made": [
                check["issue_code"] for check in criteria["conditional_content_checks"]
            ],
            "detail_witness_rule_fa": "ارجاع با شاهد همان بازو سنجیده شود. شمارهٔ منابع بازوی منتخب به بازوی پایه منتقل نشود.",
        },
        "baseline_request_receipt": baseline_ref,
        "baseline_request_payload_sha256": baseline_request_sha,
        "selected_request_payload_sha256": prepared["request_payload_sha256"],
        "historical_request_payload_sha256": historical["prior_request_payload_sha256"],
        "baseline_payload_matches_historical_dispatch": historical_match,
        "baseline_payload_matches_selected_dispatch": same_as_selected,
        "baseline_top10_overlap_with_selected": len(set(baseline_ids) & set(selected_ids)),
        "baseline_response_candidate": (
            historical["prior_response_receipt_path"]
            if historical_match and not same_as_selected else None
        ),
        "required_new_output_for_case_comparison": (
            None if historical_match or same_as_selected
            else "ONE_DEEPSEEK_RESPONSE_WITH_RECEIPT_BOUND_TO_C172_BASELINE_PAYLOAD"
        ),
        "formal_baseline_status": formal["component_comparison_status"],
        "formal_baseline_reopen_condition": formal["comparison_plan"]["required_before_dispatch"],
        "live_end_to_end_regression_verified": False,
        "new_api_calls": 0,
        "product_regression_tests_executed": 0,
        "done_item_count": formal["done_item_count"],
        "open_item_count": formal["open_item_count"],
        "selected_version_gate": formal["selected_version_gate"],
        "next_product_dependency": next_step,
    }
    result["output"] = save("C172.R11_savings_controlled_comparison_plan", result)
    print(json.dumps(result, ensure_ascii=False, indent=2))
    return result


V3_C172 = run_v3_c172()
````

این سل هش فایل‌های طرح و درخواست پایهٔ C172 را بررسی می‌کند. اگر رسید و پاسخ همان درخواست در مسیر تعیین‌شده موجود باشند، پس از تطبیق هش‌ها، متن پاسخ را برای داوری آماده می‌کند. اگر هر دو فایل غایب باشند، ثبت می‌کند که برای تکمیل مقایسه دقیقاً یک پاسخ DeepSeek برای درخواست پایهٔ C172 لازم است.

**معیار داوری هر بازو**

- هستهٔ پاسخ باید مستقیماً به سؤال جواب دهد و در شواهد همان بازو پشتوانه داشته باشد.
- اگر پاسخ بازهٔ سه تا پنج سال برای سن ۱۲ تا ۱۵ را بیان کند، ارجاع باید به شاهد صریح همان بازو برسد.
- اگر پاسخ پیشنهاد ۲۰/۸۰ را بیان کند، باید روشن باشد که این پیشنهاد دربارهٔ پول‌توجیبی کودکان است.
- حذف این دو جزئیات اختیاری به‌تنهایی خطا محسوب نمی‌شود.

این سل آفلاین است و دامنهٔ مقایسه به یک سناریو با ورودی ثابت محدود است. وضعیت بند ۱۱٫۳ همچنان **۷ مورد انجام‌شده و ۵ مورد باز** است و پذیرش نسخهٔ منتخب مسدود می‌ماند.

</div>

````python
# C173 — R11.3 — Verify the controlled baseline request and check for its response

def run_v3_c173():
    def require(condition, code):
        if not condition:
            raise ValueError(code)

    def sha(data):
        return hashlib.sha256(data).hexdigest()

    reports = Path("reports/phase2/quality_improvement/NB22E.lab.v3")
    plan_sha = "47f38d72563dbd49dd69ffdced484ea6a27b294d6326a5836863b4054e615587"
    request_file_sha = "6b79dd07856f785c29aaddcc21538a90c77b49022c30f13e385a5a0dc07d360f"
    payload_sha = "177d16c81f0251b5475e8b22892d960d96a68814b0ef4f16d76fc86626e4b799"

    root = next(
        (p for p in (Path.cwd(), *Path.cwd().parents) if (p / reports).is_dir()),
        None,
    )
    require(root is not None, "C173_PROJECT_ROOT_NOT_FOUND")

    def read_bound(relative_path, expected_sha):
        raw = (root / relative_path).read_bytes()
        require(sha(raw) == expected_sha, f"C173_HASH_MISMATCH:{relative_path}")
        return json.loads(raw)

    plan_path = reports / f"C172.R11_savings_controlled_comparison_plan.{plan_sha}.json"
    request_path = reports / f"C172.R11_savings_controlled_baseline_request.{request_file_sha}.json"
    plan = read_bound(plan_path, plan_sha)
    request_record = read_bound(request_path, request_file_sha)

    require(isinstance(request_record, dict), "C173_REQUEST_RECORD_INVALID")
    require(
        plan["baseline_request_receipt"]["path"] == request_path.as_posix()
        and plan["baseline_request_receipt"]["file_bytes_sha256"] == request_file_sha
        and plan["baseline_request_payload_sha256"] == payload_sha,
        "C173_BASELINE_REQUEST_BINDING_CONFLICT",
    )
    require(
        len({
            payload_sha,
            plan["selected_request_payload_sha256"],
            plan["historical_request_payload_sha256"],
        }) == 3,
        "C173_EXPECTED_DISTINCT_REQUESTS",
    )

    identity = {
        key: plan[key]
        for key in ("candidate_id", "question_sha256", "regression_spec_id")
    }
    response_dir = reports / "C173.R11_savings_baseline_receipts"
    receipt_path = response_dir / f"{payload_sha}.receipt.json"
    response_path = response_dir / f"{payload_sha}.response.json"

    result = {
        "cell": "C173",
        "roadmap_action": "R11",
        "roadmap_subtask": "R11.3",
        **identity,
        "input_file_hashes": {"C172": plan_sha, "C172_baseline_request": request_file_sha},
        "comparison_scope": plan["comparison_scope"],
        "fixed": plan["fixed"],
        "variable": plan["variable"],
        "case_criteria": plan["case_criteria"],
        "baseline_request_payload_sha256": payload_sha,
        "baseline_request_receipt": plan["baseline_request_receipt"],
        "expected_baseline_response_files": {
            "receipt": receipt_path.as_posix(),
            "raw_response": response_path.as_posix(),
        },
        "required_receipt_fields": [
            *identity, "request_file_bytes_sha256", "request_payload_sha256",
            "response_file_bytes_sha256",
        ],
        "baseline_response_bound": False,
        "missing_baseline_output_count": 1,
        "status": "ONE_CONTROLLED_BASELINE_RESPONSE_REQUIRED",
        "formal_baseline_status": plan["formal_baseline_status"],
        "live_end_to_end_regression_verified": False,
        "product_regression_tests_executed": 0,
        "new_api_calls": 0,
        "done_item_count": 7,
        "open_item_count": 5,
        "selected_version_gate": "BLOCKED",
        "next_product_dependency": "DISPATCH_ONLY_C172_BASELINE_REQUEST_AND_STORE_BOUND_RESPONSE",
    }

    receipt_exists = (root / receipt_path).is_file()
    response_exists = (root / response_path).is_file()
    if receipt_exists and response_exists:
        receipt_raw = (root / receipt_path).read_bytes()
        receipt = json.loads(receipt_raw)
        require(
            all(receipt.get(key) == value for key, value in identity.items())
            and receipt.get("request_file_bytes_sha256") == request_file_sha
            and receipt.get("request_payload_sha256") == payload_sha,
            "C173_RESPONSE_RECEIPT_REQUEST_MISMATCH",
        )
        response = read_bound(response_path, receipt["response_file_bytes_sha256"])
        choices = response.get("choices")
        require(
            isinstance(choices, list) and len(choices) == 1,
            "C173_EXPECTED_ONE_RESPONSE_CHOICE",
        )
        answer = choices[0].get("message", {}).get("content")
        require(
            choices[0].get("finish_reason") == "stop"
            and isinstance(answer, str) and bool(answer.strip())
            and str(response.get("model", "")).lower() == "deepseek-v4-flash",
            "C173_BASELINE_RESPONSE_INCOMPLETE_OR_WRONG_MODEL",
        )
        result.update(
            status="BASELINE_RESPONSE_BOUND_READY_FOR_PAIRED_REVIEW",
            baseline_response_bound=True,
            missing_baseline_output_count=0,
            baseline_response_receipt_sha256=sha(receipt_raw),
            baseline_response_text_sha256=sha(answer.encode("utf-8")),
            baseline_answer_for_review=answer,
            next_product_dependency="REVIEW_BOTH_ANSWERS_AGAINST_THEIR_OWN_ORDERED_EVIDENCE",
        )
    elif receipt_exists or response_exists:
        result.update(
            status="BASELINE_RESPONSE_ARTIFACT_PAIR_INCOMPLETE",
            next_product_dependency="RECOVER_MISSING_RESPONSE_ARTIFACT_BEFORE_ANY_RESEND",
        )

    data = (json.dumps(result, ensure_ascii=False, indent=2, sort_keys=True) + "\n").encode("utf-8")
    digest = sha(data)
    output = reports / f"C173.R11_savings_baseline_readiness.{digest}.json"
    output_file = root / output
    if not output_file.exists():
        output_file.write_bytes(data)
    require(output_file.read_bytes() == data, "C173_OUTPUT_RELOAD_MISMATCH")

    result["output"] = {
        "path": output.as_posix(),
        "file_bytes_sha256": digest,
        "reload_validation": "PASS",
    }
    print(json.dumps(result, ensure_ascii=False, indent=2))
    return result


V3_C173 = run_v3_c173()
````

### سل C174 — دریافت یک پاسخ برای بازوی پایهٔ سؤال پس‌انداز

این سل ابتدا فایل درخواست C172 و هش بدنهٔ درخواست را بررسی می‌کند. سپس فقط
همان بدنه را با مدل و پارامترهای ثبت‌شده به همان درگاه C157 می‌فرستد.
پاسخ خام و رسید اتصال آن به درخواست در مسیرهای تعیین‌شده در C173 ذخیره
می‌شوند و همان C173 آن‌ها را برای داوری جفتی اعتبارسنجی می‌کند.

اگر پاسخ و رسید از قبل موجود باشند، سل از همان‌ها استفاده می‌کند. اگر
اجرای قبلی در وضعیت نامشخص مانده باشد، سل پیش از ارسال مجدد متوقف می‌شود.

دامنهٔ این گام یک سؤال با ورودی ثابت است. پذیرش نسخهٔ منتخب همچنان مسدود است.

</div>

````python
# C174 — R11.3 — Send only the pinned DeepSeek baseline request

def run_v3_c174():
    from urllib.error import HTTPError, URLError
    from urllib.request import HTTPRedirectHandler, Request, build_opener

    def require(ok, code):
        if not ok:
            raise ValueError(code)

    def sha(raw):
        return hashlib.sha256(raw).hexdigest()

    def json_bytes(value):
        return (json.dumps(value, ensure_ascii=False, indent=2) + "\n").encode("utf-8")

    def write_once(path, raw):
        with path.open("xb") as stream:
            stream.write(raw)
            stream.flush()
            os.fsync(stream.fileno())

    reports = Path("reports/phase2/quality_improvement/NB22E.lab.v3")
    cwd = Path.cwd()
    root = next((p for p in (cwd, *cwd.parents) if (p / reports).is_dir()), None)
    require(root is not None, "C174_PROJECT_ROOT_NOT_FOUND")
    require(callable(globals().get("run_v3_c173")), "C174_RUN_C173_FIRST")

    def read_bound(relative, digest):
        raw = (root / relative).read_bytes()
        require(sha(raw) == digest, f"C174_FILE_HASH_MISMATCH:{relative}")
        return json.loads(raw)

    readiness_sha = "5fc5bc3ae083231c9631f209e766b1caeff855c7db226262d02a2abf24bc4502"
    request_sha = "6b79dd07856f785c29aaddcc21538a90c77b49022c30f13e385a5a0dc07d360f"
    payload_sha = "177d16c81f0251b5475e8b22892d960d96a68814b0ef4f16d76fc86626e4b799"

    plan = read_bound(
        reports / f"C173.R11_savings_baseline_readiness.{readiness_sha}.json",
        readiness_sha,
    )
    request_pin = plan["baseline_request_receipt"]
    require(
        plan["status"] == "ONE_CONTROLLED_BASELINE_RESPONSE_REQUIRED"
        and plan["baseline_request_payload_sha256"] == payload_sha
        and request_pin["file_bytes_sha256"] == request_sha,
        "C174_BASELINE_PLAN_MISMATCH",
    )

    files = plan["expected_baseline_response_files"]
    receipt = root / files["receipt"]
    response_file = root / files["raw_response"]
    attempt = receipt.with_name(f"{payload_sha}.attempt.json")
    if receipt.is_file() and response_file.is_file():
        return run_v3_c173()

    require(
        not any(p.exists() for p in (receipt, response_file, attempt)),
        "C174_PREVIOUS_ATTEMPT_OR_PARTIAL_ARTIFACT_RECOVER_BEFORE_RESEND",
    )

    def candidate_bodies(value):
        if isinstance(value, dict):
            if {"model", "messages"} <= value.keys():
                for sort_keys in (False, True):
                    for separators in (None, (",", ":")):
                        for ensure_ascii in (False, True):
                            yield json.dumps(
                                value, sort_keys=sort_keys,
                                separators=separators, ensure_ascii=ensure_ascii,
                            ).encode("utf-8")
            for child in value.values():
                yield from candidate_bodies(child)
        elif isinstance(value, list):
            for child in value:
                yield from candidate_bodies(child)
        elif isinstance(value, str):
            yield value.encode("utf-8")

    record = read_bound(request_pin["path"], request_sha)
    body = next((raw for raw in candidate_bodies(record) if sha(raw) == payload_sha), None)
    require(body is not None, "C174_EXACT_REQUEST_BYTES_NOT_RECOVERED")
    payload = json.loads(body)
    require(
        payload.get("model") == "DeepSeek-V4-Flash"
        and payload.get("temperature") == 0
        and payload.get("max_tokens") == 600
        and payload.get("thinking") == {"type": "disabled"},
        "C174_REQUEST_PROFILE_MISMATCH",
    )

    gateway = os.environ.get("ARVAN_GATEWAY_URL", "").strip().rstrip("/")
    require(gateway == "https://ai-diag.ragfinance.ir/v1", "C174_GATEWAY_MISSING_OR_MISMATCH")
    api_key = os.environ.get("ARVAN_GATEWAY_API_KEY", "").strip()
    require(bool(api_key), "C174_ARVAN_GATEWAY_API_KEY_NOT_IN_KERNEL_ENV")

    class NoRedirect(HTTPRedirectHandler):
        def redirect_request(self, req, fp, code, msg, headers, newurl):
            return None

    request = Request(
        gateway + "/chat/completions",
        data=body,
        headers={
            "Authorization": f"apikey {api_key}",
            "Content-Type": "application/json",
        },
        method="POST",
    )

    binding = {
        key: plan[key]
        for key in ("candidate_id", "question_sha256", "regression_spec_id")
    }
    binding.update(
        request_file_bytes_sha256=request_sha,
        request_payload_sha256=payload_sha,
    )
    receipt.parent.mkdir(parents=True, exist_ok=True)
    write_once(attempt, json_bytes({**binding, "maximum_dispatch_attempts": 1}))

    try:
        with build_opener(NoRedirect()).open(request, timeout=180) as reply:
            http_status = reply.status
            raw = reply.read()
    except HTTPError as error:
        raise RuntimeError(f"C174_HTTP_STATUS:{error.code}") from None
    except (URLError, TimeoutError) as error:
        raise RuntimeError(f"C174_DISPATCH_OUTCOME_UNKNOWN:{type(error).__name__}") from None

    write_once(response_file, raw)
    require(http_status == 200, f"C174_HTTP_STATUS:{http_status}")
    parsed = json.loads(raw)
    choices = parsed.get("choices")
    require(isinstance(choices, list) and len(choices) == 1, "C174_EXPECTED_ONE_CHOICE")
    choice = choices[0]
    answer = choice.get("message", {}).get("content")
    require(
        str(parsed.get("model", "")).lower() == "deepseek-v4-flash"
        and choice.get("finish_reason") == "stop"
        and isinstance(answer, str)
        and bool(answer.strip()),
        "C174_RESPONSE_INCOMPLETE_OR_WRONG_MODEL",
    )

    write_once(receipt, json_bytes({
        **binding,
        "response_file_bytes_sha256": sha(raw),
        "response_text_sha256": sha(answer.encode("utf-8")),
        "response_model": parsed["model"],
        "http_status": http_status,
        "usage_as_reported": parsed.get("usage"),
        "new_api_calls_attempted": 1,
    }))
    return run_v3_c173()


V3_C174 = run_v3_c174()
````

### سل C175 — آماده‌سازی شواهد داوری جفتی سؤال پس‌انداز

این سل پاسخ‌های بازوی پایه و بازوی منتخب را همراه با متن دقیق ارسالی به
مدل در همان بازو نمایش می‌دهد. فایل‌های ورودی و متن پاسخ‌ها با هش بررسی
می‌شوند.

در داوری، ابتدا پشتوانهٔ هستهٔ پاسخ بررسی می‌شود. سپس ادعاهای تکمیلی
و ارجاع‌های هر پاسخ با شمارهٔ منابع همان بازو سنجیده می‌شوند. دو معیار
شرطی C167 فقط در صورت بیان ادعای مربوط اعمال می‌شوند.

این سل آفلاین است. نتیجهٔ داوری به همین یک سؤال محدود است و وضعیت
پذیرش نسخهٔ منتخب همچنان `BLOCKED` است.

</div>

````python
# C175 — R11.3 — Prepare both answers and their dispatched evidence for review

def run_v3_c175():
    def require(condition, code):
        if not condition:
            raise ValueError(code)

    def sha(raw):
        return hashlib.sha256(raw).hexdigest()

    reports = Path("reports/phase2/quality_improvement/NB22E.lab.v3")
    cwd = Path.cwd()
    root = next((p for p in (cwd, *cwd.parents) if (p / reports).is_dir()), None)
    require(root is not None, "C175_PROJECT_ROOT_NOT_FOUND")

    def read_bound(relative, digest):
        raw = (root / relative).read_bytes()
        require(sha(raw) == digest, f"C175_FILE_HASH_MISMATCH:{relative}")
        return json.loads(raw)

    def read_report(label, digest):
        return read_bound(reports / f"{label}.{digest}.json", digest)

    baseline = read_report(
        "C173.R11_savings_baseline_readiness",
        "c6fc86037d93246a1e4b049a70439e6bedf17b1d28ceec726904430254737a52",
    )
    selected_plan = read_report(
        "C156.R11_target_deepseek_request_preparation",
        "ec4cd70abcd81fdae2659620da30edcf108d940df1e8a1bbed78fb5d7224730d",
    )
    selected = read_report(
        "C157.R11_target_deepseek_execution",
        "d180883a9d939626fe5edd238dd09a3ed5bbf3512c55e5b38dee9e7dd01e29bf",
    )

    identity = {
        key: baseline[key]
        for key in ("candidate_id", "question_sha256", "regression_spec_id")
    }
    require(
        baseline["baseline_response_bound"] is True
        and all(
            record.get(key) == value
            for record in (selected_plan, selected)
            for key, value in identity.items()
        ),
        "C175_SCENARIO_BINDING_MISMATCH",
    )

    baseline_digest = baseline["baseline_request_payload_sha256"]
    selected_digest = selected_plan["request_payload_sha256"]
    selected_request_pin = selected_plan["request_receipt"]
    require(
        baseline_digest != selected_digest
        and selected["request_payload_sha256"] == selected_digest
        and selected["input_file_hashes"]["C156_request"]
        == selected_request_pin["file_bytes_sha256"],
        "C175_REQUEST_BINDING_MISMATCH",
    )

    baseline_pin = baseline["baseline_request_receipt"]
    baseline_files = baseline["expected_baseline_response_files"]
    baseline_receipt = read_bound(
        baseline_files["receipt"], baseline["baseline_response_receipt_sha256"]
    )
    require(
        all(baseline_receipt.get(key) == value for key, value in identity.items())
        and baseline_receipt["request_payload_sha256"] == baseline_digest
        and baseline_receipt["request_file_bytes_sha256"] == baseline_pin["file_bytes_sha256"],
        "C175_BASELINE_RESPONSE_RECEIPT_MISMATCH",
    )

    baseline_response = read_bound(
        baseline_files["raw_response"], baseline_receipt["response_file_bytes_sha256"]
    )
    choices = baseline_response.get("choices")
    require(
        isinstance(choices, list) and len(choices) == 1
        and choices[0].get("finish_reason") == "stop"
        and str(baseline_response.get("model", "")).lower() == "deepseek-v4-flash",
        "C175_BASELINE_RESPONSE_PROFILE_MISMATCH",
    )
    baseline_answer = choices[0].get("message", {}).get("content")
    require(
        isinstance(baseline_answer, str) and bool(baseline_answer.strip())
        and sha(baseline_answer.encode("utf-8")) == baseline["baseline_response_text_sha256"],
        "C175_BASELINE_ANSWER_HASH_MISMATCH",
    )

    require(
        selected.get("status") == "TARGET_DEEPSEEK_RESPONSE_BOUND_FOR_REVIEW"
        and selected.get("http_status") == 200
        and str(selected.get("response_model", "")).lower() == "deepseek-v4-flash"
        and selected.get("finish_reason") == "stop",
        "C175_SELECTED_EXECUTION_PROFILE_MISMATCH",
    )
    selected_pin = selected["response_receipt"]
    selected_record = read_bound(selected_pin["path"], selected_pin["file_bytes_sha256"])

    def walk_values(value):
        yield value
        if isinstance(value, dict):
            for child in value.values():
                yield from walk_values(child)
        elif isinstance(value, list):
            for child in value:
                yield from walk_values(child)

    def find_answer(record, digest):
        for value in walk_values(record):
            if not isinstance(value, str) or not value.strip():
                continue
            if sha(value.encode("utf-8")) == digest:
                return value
            # Some receipts store the raw response body as a JSON string.
            if value.lstrip().startswith(("{", "[")):
                try:
                    nested = json.loads(value)
                except ValueError:
                    continue
                answer = find_answer(nested, digest)
                if answer is not None:
                    return answer
        return None

    selected_answer = find_answer(selected_record, selected["response_text_sha256"])
    require(
        selected_answer is not None,
        "C175_SELECTED_ANSWER_TEXT_NOT_FOUND_IN_C157_RECEIPT",
    )

    def load_payload(pin, digest):
        record = read_bound(pin["path"], pin["file_bytes_sha256"])
        for value in walk_values(record):
            if isinstance(value, str):
                if sha(value.encode("utf-8")) == digest:
                    return json.loads(value)
                continue
            if not isinstance(value, dict) or not {"model", "messages"} <= value.keys():
                continue
            for sort_keys in (False, True):
                for separators in (None, (",", ":")):
                    for ensure_ascii in (False, True):
                        raw = json.dumps(
                            value,
                            sort_keys=sort_keys,
                            separators=separators,
                            ensure_ascii=ensure_ascii,
                        ).encode("utf-8")
                        if sha(raw) == digest:
                            return json.loads(raw)
        raise ValueError(f"C175_EXACT_REQUEST_BYTES_NOT_RECOVERED:{digest}")

    baseline_payload = load_payload(baseline_pin, baseline_digest)
    selected_payload = load_payload(selected_request_pin, selected_digest)
    require(
        {k: v for k, v in baseline_payload.items() if k != "messages"}
        == {k: v for k, v in selected_payload.items() if k != "messages"},
        "C175_GENERATION_PARAMETERS_DIFFER",
    )

    baseline_messages = baseline_payload["messages"]
    selected_messages = selected_payload["messages"]
    require(
        [m["role"] for m in baseline_messages]
        == [m["role"] for m in selected_messages]
        == ["system", "user"]
        and baseline_messages[0] == selected_messages[0]
        and baseline_messages[1]["content"] != selected_messages[1]["content"],
        "C175_MESSAGE_PAIR_MISMATCH",
    )

    result = {
        "cell": "C175",
        "status": "PAIRED_EVIDENCE_READY_FOR_SEMANTIC_REVIEW",
        **identity,
        "case_criteria": baseline["case_criteria"],
        "arms": {
            "baseline_reranker": {
                "request_payload_sha256": baseline_digest,
                "answer": baseline_answer,
                "dispatched_user_message": baseline_messages[1]["content"],
            },
            "selected_char_rrf": {
                "request_payload_sha256": selected_digest,
                "answer": selected_answer,
                "dispatched_user_message": selected_messages[1]["content"],
            },
        },
        "paired_verdict": None,
        "formal_baseline_status": baseline["formal_baseline_status"],
        "new_api_calls": 0,
        "product_regression_tests_executed": 0,
        "done_item_count": baseline["done_item_count"],
        "open_item_count": baseline["open_item_count"],
        "selected_version_gate": baseline["selected_version_gate"],
    }

    summary = {key: value for key, value in result.items() if key != "arms"}
    print(json.dumps(summary, ensure_ascii=False, indent=2))
    for name, arm in result["arms"].items():
        print(f"\n--- {name} | ANSWER ---\n{arm['answer']}")
        print(f"\n--- {name} | DISPATCHED USER MESSAGE ---\n{arm['dispatched_user_message']}")

    return result


V3_C175 = run_v3_c175()
````

### سل C176 — ثبت رأی جفتی سؤال پس‌انداز

هستهٔ هر دو پاسخ با شواهد همان بازو پشتیبانی می‌شود. پاسخ منتخب در ارجاع
بازهٔ سنی و دامنهٔ پیشنهاد ۲۰/۸۰ ایراد دارد. پاسخ پایه این دو ادعا را مطرح
نکرده و در توصیهٔ تکمیلی دربارهٔ صندوق درآمد ثابت از عبارت منبع فراتر رفته است.

این سل رأی بررسی‌شده را به هش پاسخ‌ها و پیام‌های ارسالی C175 متصل می‌کند.
نتیجه در دامنهٔ یک سؤال با ورودی ثابت ثبت می‌شود و وضعیت پذیرش رسمی

````python
# NB22E-C183 — Exercise the computed rank chain with live DeepSeek generation.
# Persist the input before one dispatch and bind the live response to that invocation.

def run_v3_c183():
    def require(ok, code):
        if not ok:
            raise ValueError(code)

    def encode(value):
        return json.dumps(
            value, ensure_ascii=False, sort_keys=True,
            separators=(",", ":"), allow_nan=False,
        ).encode("utf-8")

    def sha(raw):
        return hashlib.sha256(raw).hexdigest()

    reports = Path("reports/phase2/quality_improvement/NB22E.lab.v3")
    cwd = Path.cwd()
    root = next(
        (p.resolve() for p in (cwd, *cwd.parents) if (p / reports).is_dir()),
        None,
    )
    require(root is not None, "C183_PROJECT_ROOT_NOT_FOUND")

    def read_bound(relative, digest):
        path = (root / relative).resolve()
        require(path.is_relative_to(root), "C183_PATH_OUTSIDE_PROJECT")
        raw = path.read_bytes()
        require(sha(raw) == digest, f"C183_HASH_MISMATCH:{relative}")
        return json.loads(raw)

    def read_report(stem, digest):
        return read_bound(reports / f"{stem}.{digest}.json", digest)

    def write_once(path, raw):
        with path.open("xb") as stream:
            stream.write(raw)
            stream.flush()
            os.fsync(stream.fileno())
        require(path.read_bytes() == raw, "C183_WRITE_RELOAD_MISMATCH")

    c182_sha = "5dc5599a75fdedced7a4e91c61669ea13edb3fb9050d9fe1fcbc3b0f2debccb2"
    state = read_report("C182.R11_item11_product_invocation_evidence", c182_sha)
    require(
        state["status"] == "PRODUCT_RANK_AND_REPLAY_VERIFIED_LIVE_SAME_CALL_OPEN"
        and state["open_item_ids"] == [9, 10, 11, 12],
        "C183_SUCCESSFUL_C182_REQUIRED",
    )
    identity = {
        key: state[key]
        for key in ("candidate_id", "question_sha256", "regression_spec_id")
    }

    c156_sha = "ec4cd70abcd81fdae2659620da30edcf108d940df1e8a1bbed78fb5d7224730d"
    prepared = read_report("C156.R11_target_deepseek_request_preparation", c156_sha)
    pin = prepared["request_receipt"]
    job = read_bound(pin["path"], pin["file_bytes_sha256"])
    request = job["request"]
    body = encode(request)
    request_sha = sha(body)

    require(
        all(doc[key] == value for doc in (prepared, job) for key, value in identity.items())
        and request_sha == prepared["request_payload_sha256"]
        == job["request_payload_sha256"]
        and request_sha == "55337f177e377bbdfcd04a163466a8ae5e1ac0944267c06c2dcd482da8d4df9f"
        and request["model"] == "DeepSeek-V4-Flash"
        and request["temperature"] == 0
        and request["max_tokens"] == 600
        and request["thinking"] == {"type": "disabled"}
        and [m["role"] for m in request["messages"]] == ["system", "user"],
        "C183_PINNED_REQUEST_MISMATCH",
    )

    c042_sha = "e8ff5073b9a0c595dda7a0cd7d49975518dac17c03c9130b9ae3c3cd35b60e8b"
    tariff = read_report("C042.R11_deepseek_batch", c042_sha)
    input_rate = tariff["input_toman_per_million"]
    output_rate = tariff["output_toman_per_million"]
    require(
        all(type(rate) is int and rate > 0 for rate in (input_rate, output_rate)),
        "C183_HISTORICAL_RATE_RECORD_INVALID",
    )
    estimate = ((len(body) + 1024) * input_rate + 600 * output_rate) / 1_000_000
    require(estimate < 100_000, "C183_HISTORICAL_ESTIMATE_REACHES_BUDGET")

    parser = globals().get("dotenv_values")
    url_module = globals().get("urllib")
    require(
        callable(parser) and url_module is not None
        and all(hasattr(url_module, name) for name in ("parse", "request", "error")),
        "C183_C001_DOTENV_AND_URLLIB_REQUIRED",
    )
    config = parser(root / ".env", interpolate=False)
    base = str(os.environ.get("ARVAN_GATEWAY_URL") or config.get("ARVAN_GATEWAY_URL") or "").strip()
    key = str(os.environ.get("ARVAN_GATEWAY_API_KEY") or config.get("ARVAN_GATEWAY_API_KEY") or "").strip()
    url = url_module.parse.urlsplit(base)
    require(
        url.scheme == "https" and bool(url.hostname) and bool(key)
        and url.username is None and url.password is None
        and not url.query and not url.fragment,
        "C183_INVALID_GATEWAY_CONFIGURATION",
    )
    route = url.path.rstrip("/") or "/v1"
    if not route.endswith("/chat/completions"):
        route += "/chat/completions"
    endpoint = url._replace(path=route).geturl()

    # Recover definitions without executing notebook cells.
    notebook_path = root / "notebooks/phase2/22e_retrieval_experiment_lab_v3.ipynb"
    notebook = json.loads(notebook_path.read_bytes())
    history = [
        "".join(cell["source"])
        for cell in notebook["cells"] if cell["cell_type"] == "code"
    ]
    history.extend(item for item in globals().get("In", []) if isinstance(item, str))

    recovered = globals().get("C163_RECOVERED_SOURCE")
    if isinstance(recovered, dict):
        history.extend(item["source"] for item in recovered.get("definitions", []))

    def definition(name):
        header = f"def {name}():"
        for entry in reversed(history):
            if header not in entry.splitlines():
                continue
            nodes = [
                node for node in ast.parse(entry).body
                if isinstance(node, ast.FunctionDef) and node.name == name
            ]
            require(len(nodes) == 1, f"C183_AMBIGUOUS_DEFINITION:{name}")
            return ast.get_source_segment(entry, nodes[0])
        raise ValueError(f"C183_SOURCE_REQUIRED:{name}")

    original = definition("run_v3_c163")
    adapter_source = definition("run_v3_c162")
    for name in ("run_v3_c126", "run_v3_c148", "run_v3_c149", "run_v3_c150"):
        definition(name)

    def replace_once(source, old, new, label):
        require(source.count(old) == 1, f"C183_SOURCE_CONTRACT_CHANGED:{label}")
        return source.replace(old, new, 1)

    original = replace_once(
        original,
        "        template = replace_once(template, old, new, old)",
        '        require(template.count(old) == 2, "C183_EVENT_COUNT_MISMATCH")\n'
        "        template = template.replace(old, new)",
        "historical_event_loop",
    )
    original = replace_once(
        original,
        '    adapter = definition("run_v3_c162")',
        "    adapter = c183_adapter_source",
        "adapter_definition",
    )
    original = replace_once(
        original,
        '    scope = dict(globals())\n    scope.update(\n        In=[template],',
        '    template = c183_bind_template(template)\n'
        '    adapter = c183_bind_adapter(adapter)\n'
        '    scope = dict(globals())\n    scope.update(\n        In=[template],',
        "live_adapter_hook",
    )
    original = original.replace(
        'result.get("saved_responses_replayed") == 1',
        'result.get("live_responses_generated") == 1',
    )

    folder = root / reports / "C183.R11_live_product_receipts"
    folder.mkdir(exist_ok=True)
    started_path = folder / f"{request_sha}.started.json"
    response_path = folder / f"{request_sha}.response.json"
    require(
        not started_path.exists() and not response_path.exists(),
        "C183_PREVIOUS_ATTEMPT_EXISTS_RECOVER_BEFORE_RESEND",
    )

    invocation_id = sha(
        f"C183:{request_sha}:{os.getpid()}:{time.time_ns()}".encode("utf-8")
    )
    live = {"answer_sha": None, "dispatch_count": 0}

    def bind_template(template):
        return replace_once(
            template,
            '            "pipeline_source_sha256": pipeline_sha,',
            '            "pipeline_source_sha256": pipeline_sha,\n'
            '            "invocation_id": c183_invocation_id,',
            "input_invocation_id",
        )

    def bind_adapter(adapter):
        adapter = replace_once(
            adapter,
            "answer = c162_replay_at_boundary(question, context)",
            "answer = c183_live_at_boundary(question, context, receipt_ref)",
            "live_generation_boundary",
        )
        require(adapter.count("== c162_answer_sha") == 2,
                "C183_ANSWER_CHECK_COUNT_MISMATCH")
        adapter = adapter.replace(
            "== c162_answer_sha", '== c183_live_state["answer_sha"]'
        )
        for old, new in (
            ("C163", "C183"),
            ("saved_response_returned", "live_response_received_and_verified"),
            ("saved_responses_replayed", "live_responses_generated"),
            ("SAVED_DEEPSEEK_RESPONSE_REPLAY", "LIVE_DEEPSEEK_GENERATION"),
            ("SAVED_RESPONSE_REPLAY", "LIVE_DEEPSEEK_GENERATION"),
            ("TARGET_COMPUTED_RANK_PRODUCT_RESPONSE_REPLAY_VERIFIED",
             "TARGET_COMPUTED_RANK_PRODUCT_LIVE_RESPONSE_VERIFIED"),
            ("target_computed_rank_product_response_replay", "target_product_live_integration"),
            ("replay_input_receipt", "live_input_receipt"),
            ("replay_output_receipt", "live_output_receipt"),
            ("replayed_request_sha256", "request_payload_sha256"),
            ("replayed_response_receipt_sha256", "historical_response_reference_sha256"),
        ):
            adapter = adapter.replace(old, new)
        return adapter

    class NoRedirect(url_module.request.HTTPRedirectHandler):
        def redirect_request(self, req, fp, code, msg, headers, newurl):
            return None

    opener = url_module.request.build_opener(NoRedirect())

    def live_at_boundary(question, context, input_ref):
        require(live["dispatch_count"] == 0, "C183_DUPLICATE_GENERATION_CALL")
        recorded = read_bound(input_ref["path"], input_ref["file_bytes_sha256"])
        require(
            sha(question.encode("utf-8")) == identity["question_sha256"]
            and recorded["candidate_id"] == identity["candidate_id"]
            and recorded["question_sha256"] == identity["question_sha256"]
            and recorded["selected_variant"] == "char_rrf"
            and recorded["runtime_identity"]["invocation_id"] == invocation_id
            and recorded["rendered_context_sha256"]
            == sha(context.encode("utf-8")) == job["rendered_context_sha256"],
            "C183_PRE_DISPATCH_INPUT_MISMATCH",
        )
        user = request["messages"][1]
        require(user["content"].count(context) == 1 and question in user["content"],
                "C183_REQUEST_CONTEXT_MISMATCH")

        started = {
            **identity,
            "invocation_id": invocation_id,
            "request_payload_sha256": request_sha,
            "request_file_bytes_sha256": pin["file_bytes_sha256"],
            "input_receipt": input_ref,
            "endpoint_sha256": sha(endpoint.encode("utf-8")),
            "rate_basis": "C042_HISTORICAL_REFERENCE_RATES",
            "current_rate_verified": False,
            "historical_estimated_cost_toman": estimate,
        }
        write_once(started_path, encode(started))
        live["dispatch_count"] = 1
        outgoing = url_module.request.Request(
            endpoint, data=body, method="POST",
            headers={"Authorization": f"apikey {key}", "Content-Type": "application/json"},
        )
        try:
            with opener.open(outgoing, timeout=180) as response:
                status, raw_response = response.status, response.read()
        except url_module.error.HTTPError as error:
            with error:
                status, raw_response = error.code, error.read()
        except (OSError, ValueError):
            raise ValueError("C183_DISPATCH_OUTCOME_UNKNOWN_NO_RESEND") from None

        response_record = {
            **identity,
            "invocation_id": invocation_id,
            "request_payload_sha256": request_sha,
            "input_receipt_file_bytes_sha256": input_ref["file_bytes_sha256"],
            "started_receipt_sha256": sha(started_path.read_bytes()),
            "http_status": status,
            "response_body": raw_response.decode("utf-8"),
            "response_body_sha256": sha(raw_response),
        }
        write_once(response_path, encode(response_record))
        require(status == 200, "C183_NON200_RESPONSE_PRESERVED_NO_RESEND")
        envelope = json.loads(raw_response)
        choices = envelope["choices"]
        require(len(choices) == 1, "C183_EXPECTED_ONE_CHOICE")
        choice = choices[0]
        answer = choice["message"]["content"]
        model = envelope["model"].strip().rsplit("/", 1)[-1].lower()
        require(
            model in {"deepseek-v4-flash", "deepseek-v4-flash-0731"}
            and choice.get("finish_reason") == "stop"
            and isinstance(answer, str) and bool(answer.strip()),
            "C183_RESPONSE_MODEL_OR_FINISH_MISMATCH",
        )
        live.update(
            answer_sha=sha(answer.encode("utf-8")),
            response=answer,
            response_model=model,
            usage_as_reported=envelope.get("usage"),
            response_receipt={
                "path": response_path.relative_to(root).as_posix(),
                "file_bytes_sha256": sha(response_path.read_bytes()),
            },
        )
        return answer

    scope = dict(globals())
    scope.update(
        In=history,
        c183_adapter_source=adapter_source,
        c183_bind_template=bind_template,
        c183_bind_adapter=bind_adapter,
        c183_live_at_boundary=live_at_boundary,
        c183_live_state=live,
        c183_invocation_id=invocation_id,
    )
    exec(compile(original, "<C183:reviewed-C163-live-adapter>", "exec"), scope)
    result = scope["run_v3_c163"]()

    passed = (
        result.get("status") == "TARGET_COMPUTED_RANK_PRODUCT_LIVE_RESPONSE_VERIFIED"
        and result.get("rank_providers_executed_in_target_product_call") is True
        and live["dispatch_count"] == 1 and live["answer_sha"] is not None
        and result.get("output_receipt") is not None
    )
    result.pop("output", None)
    result.update(
        cell="C183",
        invocation_id=invocation_id,
        live_generation_inside_same_product_call_verified=passed,
        new_api_calls=live["dispatch_count"],
        live_response_receipt=live.get("response_receipt"),
        response=live.get("response"),
        response_text_sha256=live.get("answer_sha"),
        usage_as_reported=live.get("usage_as_reported"),
        rate_basis="C042_HISTORICAL_REFERENCE_RATES",
        current_rate_verified=False,
        historical_estimated_cost_toman=estimate,
        done_item_count=8,
        open_item_count=4,
        open_item_ids=[9, 10, 11, 12],
        product_regression_tests_executed=0,
        selected_version_gate="BLOCKED",
        next_product_dependency="ASSESS_C183_INTEGRATION_RECEIPTS_AND_ITEM11_ACCEPTANCE",
    )
    raw = encode(result)
    digest = sha(raw)
    relative = reports / f"C183.R11_live_product_integration.{digest}.json"
    path = root / relative
    if path.exists():
        require(path.read_bytes() == raw, "C183_REPORT_RELOAD_MISMATCH")
    else:
        write_once(path, raw)
    result["output"] = {
        "path": relative.as_posix(),
        "file_bytes_sha256": digest,
        "reload_validation": "PASS",
    }
    print(json.dumps(result, ensure_ascii=False, indent=2))
    return result


V3_C183 = run_v3_c183()
````

### NB22E-C185 — ثبت نتیجهٔ اولین سناریوی زنده در مورد ۱۲

````python
# NB22E-C185 — Record one live selected-version scenario result for item 12.
# Bind the expected behavior to verified runtime evidence in a separate ledger overlay.

def run_v3_c185():
    def require(ok, code):
        if not ok:
            raise ValueError(code)

    def sha(raw):
        return hashlib.sha256(raw).hexdigest()

    reports = Path("reports/phase2/quality_improvement/NB22E.lab.v3")
    cwd = Path.cwd()
    root = next(
        (p.resolve() for p in (cwd, *cwd.parents) if (p / reports).is_dir()),
        None,
    )
    require(root is not None, "C185_PROJECT_ROOT_NOT_FOUND")

    def load(relative, digest):
        path = (root / relative).resolve()
        require(path.is_relative_to(root), "C185_PATH_OUTSIDE_PROJECT")
        raw = path.read_bytes()
        require(sha(raw) == digest, f"C185_HASH_MISMATCH:{relative}")
        return json.loads(raw)

    c043_sha = "811af55fabd63d695bd2767c762055c4997d81e4acd7c36f7bdde717421e6145"
    c184_sha = "6ca4fddabbb13636b6a3eb694028c738cc88c781dd3c2051358a144ef3405c77"
    source = load(
        reports / f"C043.R11_scenario_review_inputs.{c043_sha}.json",
        c043_sha,
    )
    accepted = load(
        reports / f"C184.R11_item11_live_integration_acceptance.{c184_sha}.json",
        c184_sha,
    )

    scenarios = source["scenario_definitions"]["selected_version_execution_requests"]
    ids = [row["regression_spec_id"] for row in scenarios]
    require(
        len(ids) == len(set(ids)) == 36,
        "C185_SCENARIO_POPULATION_MISMATCH",
    )
    target_id = accepted["regression_spec_id"]
    matches = [row for row in scenarios if row["regression_spec_id"] == target_id]
    require(len(matches) == 1, "C185_TARGET_SCENARIO_NOT_UNIQUE")
    spec = matches[0]
    require(
        spec["candidate_id"] == accepted["candidate_id"]
        and spec["kind"] == "ANSWER_DECISION"
        and spec["behavior_expectation"]["expected_behavior"] == "ANSWER_SUPPORTED"
        and accepted["status"] == "ITEM_11_ONE_TARGET_LIVE_PRODUCT_INTEGRATION_VERIFIED"
        and accepted["product_runtime_binding_verified_for_this_scenario"] is True
        and accepted["live_generation_inside_same_product_call_verified"] is True
        and accepted["scoped_semantic_review"]["verdict"]
        == "ANSWER_SUPPORTED_IN_DISPATCHED_CONTEXT"
        and all(accepted["binding_checks"].values())
        and accepted["open_item_ids"] == [9, 10, 12],
        "C185_EXPECTATION_OR_EXECUTION_MISMATCH",
    )

    c183_sha = accepted["input_file_hashes"]["C183"]
    execution = load(
        reports / f"C183.R11_live_product_integration.{c183_sha}.json",
        c183_sha,
    )
    require(
        execution["regression_spec_id"] == target_id
        and execution["candidate_id"] == accepted["candidate_id"]
        and execution["question_sha256"] == accepted["question_sha256"]
        and execution["invocation_id"] == accepted["invocation_id"]
        and execution["response_text_sha256"] == accepted["response_text_sha256"]
        and sha(execution["response"].encode("utf-8"))
        == accepted["response_text_sha256"],
        "C185_RESPONSE_BINDING_MISMATCH",
    )
    for field in ("input_receipt", "output_receipt", "live_response_receipt"):
        pin = execution[field]
        load(pin["path"], pin["file_bytes_sha256"])

    result = {
        "cell": "C185",
        "roadmap_subtask": "R11.3",
        "checklist_item": 12,
        "status": "ONE_LIVE_SCENARIO_RESULT_ADDED_TO_REGRESSION_OVERLAY",
        "scope": "C043_36_RESPONSE_AND_CITATION_SCENARIOS",
        "input_file_hashes": {
            "C043": c043_sha,
            "C183": c183_sha,
            "C184": c184_sha,
        },
        "scenario_result": {
            "regression_spec_id": target_id,
            "candidate_id": accepted["candidate_id"],
            "question_sha256": accepted["question_sha256"],
            "invocation_id": accepted["invocation_id"],
            "response_text_sha256": accepted["response_text_sha256"],
            "expected_behavior": "ANSWER_SUPPORTED",
            "verdict": "PASS_SCOPED_EXPECTED_BEHAVIOR",
            "assessment": accepted["scoped_semantic_review"],
            "execution_receipts": {
                field: execution[field]
                for field in ("input_receipt", "output_receipt", "live_response_receipt")
            },
        },
        "scenario_count": 36,
        "scoped_live_scenarios_executed_in_this_overlay": 1,
        "scoped_live_scenarios_passed_in_this_overlay": 1,
        "remaining_scenarios_without_result_in_this_overlay": [
            scenario_id for scenario_id in ids if scenario_id != target_id
        ],
        "remaining_scenario_count_in_this_overlay": 35,
        "formal_component_nonregression_status": "NOT_COMPARABLE_NOT_PASS",
        "formal_component_dependency": "ACCEPTED_PRECHANGE_REPORT_AND_COMPARABLE_ROW_LEVEL_ARTIFACT",
        "item_12_status": "PARTIAL",
        "newly_completed_items_from_previous_open_set": [],
        "done_item_count": 9,
        "open_item_count": 3,
        "open_item_ids": [9, 10, 12],
        "selected_version_gate": "BLOCKED",
        "new_api_calls": 0,
    }

    raw = (
        json.dumps(result, ensure_ascii=False, sort_keys=True, indent=2) + "\n"
    ).encode("utf-8")
    digest = sha(raw)
    relative = reports / f"C185.R11_first_live_scenario_result.{digest}.json"
    path = root / relative
    try:
        with path.open("xb") as stream:
            stream.write(raw)
    except FileExistsError:
        pass
    require(path.read_bytes() == raw, "C185_OUTPUT_RELOAD_MISMATCH")
    result["output"] = {
        "path": relative.as_posix(),
        "file_bytes_sha256": digest,
        "reload_validation": "PASS",
    }
    print(json.dumps(result, ensure_ascii=False, indent=2))
    return result


V3_C185 = run_v3_c185()
````

### NB22E-C186 — کنترل تجمیعی ورودی ۳۵ سناریوی باقی‌مانده

فهرست باقی‌ماندهٔ C185 با مشخصات C043 تطبیق داده می‌شود. برای هر
کاندیدا، سؤال Golden، ترتیب تاریخی Hybrid100، ترتیب و امتیازهای
pool50 و وجود متن قطعه‌ها بررسی و به ردیف‌های منبع متصل می‌شوند.

سناریوهای دارای کاندیدای مشترک از یک ورودی مشترک استفاده می‌کنند.
انتظار هر سناریو جداگانه حفظ می‌شود. ایراد هر پرونده در همان ردیف
ثبت می‌شود تا اجرای بعدی بر اساس ورودی‌های قابل‌ردیابی انتخاب شود.

این سل آفلاین است. Gate نسخهٔ منتخب همچنان BLOCKED است.

</div>

````python
# NB22E-C186 — Bind historical inputs for all remaining response scenarios.
# Validate shared candidate inputs once and preserve each scenario's expectation.

def run_v3_c186():
    def require(ok, code):
        if not ok:
            raise ValueError(code)

    def sha(raw):
        return hashlib.sha256(raw).hexdigest()

    reports = Path("reports/phase2/quality_improvement/NB22E.lab.v3")
    cwd = Path.cwd()
    root = next(
        (p.resolve() for p in (cwd, *cwd.parents) if (p / reports).is_dir()),
        None,
    )
    require(root is not None, "C186_PROJECT_ROOT_NOT_FOUND")

    def read_bound(relative, digest):
        path = (root / relative).resolve()
        require(path.is_relative_to(root), "C186_PATH_OUTSIDE_PROJECT")
        raw = path.read_bytes()
        require(sha(raw) == digest, f"C186_HASH_MISMATCH:{relative}")
        return raw

    c185_sha = "33902850e52c924fd8f2d8ae63e2256586b91141b2e6ae46574671aecc59a6a0"
    c043_sha = "811af55fabd63d695bd2767c762055c4997d81e4acd7c36f7bdde717421e6145"
    prior = json.loads(read_bound(
        reports / f"C185.R11_first_live_scenario_result.{c185_sha}.json", c185_sha,
    ))
    source = json.loads(read_bound(
        reports / f"C043.R11_scenario_review_inputs.{c043_sha}.json", c043_sha,
    ))
    remaining = prior["remaining_scenarios_without_result_in_this_overlay"]
    all_specs = source["scenario_definitions"]["selected_version_execution_requests"]
    spec_index = {row["regression_spec_id"]: row for row in all_specs}
    require(
        len(spec_index) == len(all_specs) == 36
        and len(remaining) == len(set(remaining)) == 35
        and set(remaining) <= spec_index.keys(),
        "C186_SCENARIO_POPULATION_MISMATCH",
    )
    specs = [spec_index[spec_id] for spec_id in remaining]
    candidate_ids = {row["candidate_id"] for row in specs}

    pins = {
        "golden": (
            "artifacts/phase2/golden_dataset_v2/golden_dataset_v2.jsonl",
            "b9ca18e710686ddca48a25c637840b716adb306717cafd8ec7c96ce9e182e6f2",
        ),
        "hybrid": (
            "data/phase2/retrieval/hybrid_retrieval_v1/hybrid_candidates_v1.jsonl",
            "f54277680f3d20681b1585915f54bb4485ecb1fcccb15c27df0061ff0725d19a",
        ),
        "reranker": (
            "data/phase2/retrieval/reranker_v1/reranked_candidates_v1.jsonl",
            "d00285bbf3865fbd51041db1612b9fdc17e9778273ea6d136978906a07268b7a",
        ),
        "chunks": (
            "data/phase2/chunks/nb11_benchmark_v1/baseline_phase1_char_safe_v1.jsonl",
            "3891c4ecad9332ba753cd605bbccd99da9d063f64934a30ee39e495349cffb50",
        ),
    }

    inputs = {}
    for role in ("golden", "hybrid", "reranker"):
        relative, digest = pins[role]
        index = {}
        for number, line in enumerate(read_bound(relative, digest).splitlines(), 1):
            if not line.strip():
                continue
            row = json.loads(line)
            candidate = row.get("candidate_id")
            if candidate in candidate_ids:
                require(candidate not in index, f"C186_DUPLICATE_ROW:{role}:{candidate}")
                index[candidate] = {
                    "row": row,
                    "receipt": {
                        "path": relative,
                        "file_bytes_sha256": digest,
                        "line_number": number,
                        "row_sha256_excluding_line_ending": sha(line),
                    },
                }
        inputs[role] = index

    corpus = {}
    for line in read_bound(*pins["chunks"]).splitlines():
        if line.strip():
            row = json.loads(line)
            require(row["chunk_id"] not in corpus, "C186_DUPLICATE_CHUNK")
            corpus[row["chunk_id"]] = row

    def ordered_ids(row, depth):
        entries = row["candidates"]
        require(
            [entry["rank"] for entry in entries] == list(range(1, depth + 1)),
            f"C186_RANK_SEQUENCE_MISMATCH:{depth}",
        )
        ids = [entry["chunk_id"] for entry in entries]
        require(
            len(ids) == len(set(ids)) == depth
            and all(isinstance(cid, str) and cid for cid in ids),
            f"C186_CHUNK_ID_MISMATCH:{depth}",
        )
        return ids

    candidates = []
    for candidate in sorted(candidate_ids):
        entry = {"candidate_id": candidate}
        try:
            bound = {role: inputs[role][candidate] for role in inputs}
            golden, hybrid, reranker = (
                bound[role]["row"] for role in ("golden", "hybrid", "reranker")
            )
            question = golden["evaluation_content"]["question"]
            require(
                isinstance(question, str) and bool(question.strip()),
                "C186_QUESTION_MISSING",
            )
            split = hybrid["split"]
            require(
                split in {"development", "test"}
                and reranker.get("split", split) == split,
                "C186_SPLIT_MISMATCH",
            )
            hybrid_ids = ordered_ids(hybrid, 100)
            pool_ids = ordered_ids(reranker, 50)
            scores = {
                row["chunk_id"]: float(row["reranker_score"])
                for row in reranker["candidates"]
            }
            require(
                set(pool_ids) == set(hybrid_ids[:50]) == set(scores)
                and all(math.isfinite(score) for score in scores.values()),
                "C186_POOL50_CONTRACT_MISMATCH",
            )
            require(
                all(
                    cid in corpus
                    and isinstance(corpus[cid].get("content"), str)
                    and bool(corpus[cid]["content"].strip())
                    for cid in hybrid_ids
                ),
                "C186_HYBRID_CHUNK_TEXT_MISSING",
            )
            entry.update(
                status="HISTORICAL_INPUTS_BOUND",
                question=question,
                question_sha256=sha(question.encode("utf-8")),
                ranker_question_sha256=sha(question.strip().encode("utf-8")),
                split=split,
                source_rows={role: bound[role]["receipt"] for role in bound},
                hybrid100_order=hybrid_ids,
                hybrid100_text_sha256=[
                    sha(corpus[cid]["content"].encode("utf-8")) for cid in hybrid_ids
                ],
                pool50_order=pool_ids,
                pool50_scores_by_chunk_id=scores,
                selected_runtime_input_receipt_verified=False,
            )
        except (KeyError, TypeError, ValueError) as error:
            entry.update(
                status="HISTORICAL_INPUT_BINDING_BLOCKED",
                blocker=f"{type(error).__name__}:{error}",
            )
        candidates.append(entry)

    blocked = [
        {"candidate_id": row["candidate_id"], "blocker": row["blocker"]}
        for row in candidates if row["status"] != "HISTORICAL_INPUTS_BOUND"
    ]
    result = {
        "cell": "C186",
        "roadmap_subtask": "R11.3",
        "checklist_item": 12,
        "status": "REMAINING_SCENARIO_INPUT_PREFLIGHT_RECORDED",
        "input_file_hashes": {
            "C185": c185_sha,
            "C043": c043_sha,
            **{role: digest for role, (_, digest) in pins.items()},
        },
        "scenario_count": len(specs),
        "candidate_count": len(candidates),
        "bound_candidate_count": len(candidates) - len(blocked),
        "blocked_candidates": blocked,
        "scenario_specs": specs,
        "candidate_inputs": candidates,
        "next_dependency": (
            "RESOLVE_RECORDED_INPUT_BLOCKERS"
            if blocked else
            "BIND_QUERY_VECTORS_AND_GENERALIZE_VERIFIED_PROVIDERS_FOR_REMAINING_CANDIDATES"
        ),
        "rank_reconstruction_for_unverified_candidates": "DEFERRED",
        "new_api_calls": 0,
        "ranking_runs_executed": 0,
        "new_scenario_results": 0,
        "done_item_count": 9,
        "open_item_count": 3,
        "open_item_ids": [9, 10, 12],
        "selected_version_gate": "BLOCKED",
    }

    raw = (
        json.dumps(result, ensure_ascii=False, sort_keys=True, indent=2) + "\n"
    ).encode("utf-8")
    digest = sha(raw)
    relative = reports / f"C186.R11_remaining_scenario_input_preflight.{digest}.json"
    path = root / relative
    try:
        with path.open("xb") as stream:
            stream.write(raw)
    except FileExistsError:
        pass
    require(path.read_bytes() == raw, "C186_OUTPUT_RELOAD_MISMATCH")
    result["output"] = {
        "path": relative.as_posix(),
        "file_bytes_sha256": digest,
        "reload_validation": "PASS",
    }
    print(json.dumps(
        {
            key: value for key, value in result.items()
            if key not in {"candidate_inputs", "scenario_specs"}
        },
        ensure_ascii=False, indent=2,
    ))
    return result


V3_C186 = run_v3_c186()
````

### NB22E-C187 — اتصال تجمیعی بردارهای ذخیره‌شدهٔ سؤال‌ها

قرارداد Test به‌صورت مانع صریح گزارش می‌شود. این مرحله آفلاین است.

````python
# NB22E-C187 — Bind cached Test vectors for the remaining candidates.
# Rebuild the C147 cache contract and validate each candidate's exact array row.

def run_v3_c187():
    def require(ok, code):
        if not ok:
            raise ValueError(code)

    def sha(raw):
        return hashlib.sha256(raw).hexdigest()

    def canonical_sha(value):
        return sha(json.dumps(
            value, ensure_ascii=False, sort_keys=True, separators=(",", ":"),
            allow_nan=False,
        ).encode("utf-8"))

    reports = Path("reports/phase2/quality_improvement/NB22E.lab.v3")
    cwd = Path.cwd()
    root = next(
        (p.resolve() for p in (cwd, *cwd.parents) if (p / reports).is_dir()),
        None,
    )
    require(root is not None, "C187_PROJECT_ROOT_NOT_FOUND")

    def path_for(relative):
        path = (root / relative).resolve()
        require(path.is_relative_to(root), "C187_PATH_OUTSIDE_PROJECT")
        return path

    def read_bound(relative, digest):
        raw = path_for(relative).read_bytes()
        require(sha(raw) == digest, f"C187_HASH_MISMATCH:{relative}")
        return raw

    prior_sha = "5cb0fa8851d28b1fa459f01cfbf6355fc32f1db9d5f484a133685b93d6315821"
    prior = json.loads(read_bound(
        reports / f"C186.R11_remaining_scenario_input_preflight.{prior_sha}.json",
        prior_sha,
    ))
    require(
        prior["bound_candidate_count"] == prior["candidate_count"] == 33
        and prior["scenario_count"] == 35
        and not prior["blocked_candidates"],
        "C187_SUCCESSFUL_C186_REQUIRED",
    )

    hybrid_path = "data/phase2/retrieval/hybrid_retrieval_v1/hybrid_candidates_v1.jsonl"
    golden_path = "artifacts/phase2/golden_dataset_v2/golden_dataset_v2.jsonl"
    hybrid_rows = [
        json.loads(line)
        for line in read_bound(
            hybrid_path, prior["input_file_hashes"]["hybrid"]
        ).splitlines() if line.strip()
    ]
    ids = sorted(
        row["candidate_id"] for row in hybrid_rows if row["split"] == "test"
    )
    require(len(ids) == len(set(ids)) == 1149, "C187_TEST_POPULATION_MISMATCH")
    wanted = set(ids)
    questions = {}
    for line in read_bound(
        golden_path, prior["input_file_hashes"]["golden"]
    ).splitlines():
        if not line.strip():
            continue
        row = json.loads(line)
        candidate = row["candidate_id"]
        if candidate in wanted:
            text = row["evaluation_content"]["question"]
            require(
                candidate not in questions and isinstance(text, str) and text.strip(),
                f"C187_INVALID_QUESTION:{candidate}",
            )
            questions[candidate] = text.strip()
    require(set(questions) == wanted, "C187_QUESTION_POPULATION_MISMATCH")

    query_set_sha = canonical_sha([
        {"candidate_id": candidate, "question": questions[candidate]}
        for candidate in ids
    ])
    contract = {
        "split": "test",
        "query_set_sha256": query_set_sha,
        "model_id": "intfloat/multilingual-e5-small",
        "model_revision": "614241f622f53c4eeff9890bdc4f31cfecc418b3",
        "query_prefix": "query:",
        "normalize_embeddings": True,
    }
    cache_key = canonical_sha(contract)
    cache = Path("artifacts/phase2/hybrid_retrieval/query_embedding_cache_v1")
    metadata_relative = cache / "test_query_embeddings.json"
    array_relative = cache / "test_query_embeddings.npy"
    metadata_raw = path_for(metadata_relative).read_bytes()
    metadata = json.loads(metadata_raw)

    require(
        metadata["artifact_type"] == "NB16_TEST_QUERY_EMBEDDING_CACHE"
        and metadata["schema_version"] == "1.0"
        and metadata["candidate_ids"] == ids
        and metadata["cache_key"] == cache_key
        and metadata["query_count"] == 1149
        and metadata["shape"] == [1149, 384]
        and metadata["dtype"] == "float32"
        and all(metadata.get(key) == value for key, value in contract.items()),
        "C187_CACHE_METADATA_MISMATCH",
    )
    array_sha = sha(path_for(array_relative).read_bytes())
    array = np.load(path_for(array_relative), allow_pickle=False)
    require(
        array.shape == (1149, 384) and array.dtype == np.float32,
        "C187_CACHE_ARRAY_MISMATCH",
    )
    row_index = {candidate: index for index, candidate in enumerate(ids)}
    bindings, blocked = [], []

    for candidate_input in prior["candidate_inputs"]:
        candidate = candidate_input["candidate_id"]
        if candidate_input["split"] != "test":
            blocked.append({
                "candidate_id": candidate,
                "blocker": "NON_TEST_CANDIDATE_REQUIRES_ITS_OWN_CACHE_CONTRACT",
            })
            continue
        try:
            require(candidate in row_index, "C187_CANDIDATE_NOT_IN_TEST_CACHE")
            index = row_index[candidate]
            question_sha = sha(questions[candidate].encode("utf-8"))
            require(
                sha(candidate_input["question"].encode("utf-8"))
                == candidate_input["question_sha256"]
                and question_sha == candidate_input["ranker_question_sha256"],
                "C187_EMBEDDED_QUESTION_MISMATCH",
            )
            vector = array[index]
            norm = float(np.linalg.norm(vector))
            require(
                bool(np.isfinite(vector).all()) and bool(np.isclose(norm, 1.0)),
                "C187_INVALID_VECTOR",
            )
            bindings.append({
                "candidate_id": candidate,
                "split": "test",
                "question_sha256": candidate_input["question_sha256"],
                "embedded_question_sha256": question_sha,
                "cache_row_index_zero_based": index,
                "query_vector_sha256": sha(vector.tobytes()),
                "query_vector_norm": norm,
            })
        except ValueError as error:
            blocked.append({"candidate_id": candidate, "blocker": str(error)})

    require(
        sha(path_for(array_relative).read_bytes()) == array_sha
        and path_for(metadata_relative).read_bytes() == metadata_raw,
        "C187_CACHE_CHANGED_DURING_READ",
    )

    result = {
        "cell": "C187",
        "roadmap_subtask": "R11.3",
        "checklist_item": 12,
        "status": (
            "ALL_REMAINING_QUERY_VECTORS_BOUND"
            if not blocked else "QUERY_VECTOR_BINDING_PARTIAL"
        ),
        "input_file_hashes": {"C186": prior_sha},
        "cache_contract": contract,
        "cache_key": cache_key,
        "cache_files": {
            "metadata": {
                "path": metadata_relative.as_posix(),
                "file_bytes_sha256": sha(metadata_raw),
            },
            "array": {
                "path": array_relative.as_posix(),
                "file_bytes_sha256": array_sha,
            },
        },
        "candidate_count": prior["candidate_count"],
        "bound_vector_count": len(bindings),
        "blocked_candidates": blocked,
        "vector_bindings": bindings,
        "embedding_calls": 0,
        "qdrant_queries_attempted": 0,
        "new_api_calls": 0,
        "new_scenario_results": 0,
        "rank_reconstruction_for_unverified_candidates": "DEFERRED",
        "done_item_count": 9,
        "open_item_count": 3,
        "open_item_ids": [9, 10, 12],
        "selected_version_gate": "BLOCKED",
        "next_dependency": (
            "GENERALIZE_VERIFIED_PRODUCT_PROVIDERS_FOR_BOUND_CANDIDATES"
            if not blocked else "RESOLVE_RECORDED_VECTOR_BLOCKERS"
        ),
    }

    raw = (
        json.dumps(result, ensure_ascii=False, sort_keys=True, indent=2) + "\n"
    ).encode("utf-8")
    digest = sha(raw)
    relative = reports / f"C187.R11_remaining_query_vector_bindings.{digest}.json"
    path = path_for(relative)
    try:
        with path.open("xb") as stream:
            stream.write(raw)
    except FileExistsError:
        pass
    require(path.read_bytes() == raw, "C187_OUTPUT_RELOAD_MISMATCH")
    result["output"] = {
        "path": relative.as_posix(),
        "file_bytes_sha256": digest,
        "reload_validation": "PASS",
    }
    print(json.dumps(
        {key: value for key, value in result.items() if key != "vector_bindings"},
        ensure_ascii=False, indent=2,
    ))
    return result


V3_C187 = run_v3_c187()
````

### NB22E-C188 — اجرای تجمیعی کنترل Hybrid100 برای ۳۳ کاندیدا

````python
# NB22E-C188 — Compare all remaining Hybrid100 orders in one local batch.
# Reuse the pinned NB16 functions and the C187 vectors without new embeddings.

def run_v3_c188():
    def require(ok, code):
        if not ok:
            raise ValueError(code)

    def sha(raw):
        return hashlib.sha256(raw).hexdigest()

    reports = Path("reports/phase2/quality_improvement/NB22E.lab.v3")
    cwd = Path.cwd()
    root = next(
        (p.resolve() for p in (cwd, *cwd.parents) if (p / reports).is_dir()),
        None,
    )
    require(root is not None, "C188_PROJECT_ROOT_NOT_FOUND")

    def read_bound(relative, digest):
        path = (root / relative).resolve()
        require(path.is_relative_to(root), "C188_PATH_OUTSIDE_PROJECT")
        raw = path.read_bytes()
        require(sha(raw) == digest, f"C188_HASH_MISMATCH:{relative}")
        return raw

    c186_sha = "5cb0fa8851d28b1fa459f01cfbf6355fc32f1db9d5f484a133685b93d6315821"
    c187_sha = "419fbd3ebe5cc85e53a655516be3456688144e3d4208cc38caf036da037106a4"
    nb_sha = "40699ca54c57fcef40472df6b6fc42a54f74dc2e9bdc9bb40e0ccc93bb7ad6a7"
    inputs = json.loads(read_bound(
        reports / f"C186.R11_remaining_scenario_input_preflight.{c186_sha}.json",
        c186_sha,
    ))
    vectors = json.loads(read_bound(
        reports / f"C187.R11_remaining_query_vector_bindings.{c187_sha}.json",
        c187_sha,
    ))
    require(
        vectors["status"] == "ALL_REMAINING_QUERY_VECTORS_BOUND"
        and vectors["input_file_hashes"]["C186"] == c186_sha
        and vectors["bound_vector_count"] == inputs["candidate_count"] == 33,
        "C188_INPUT_BATCH_MISMATCH",
    )

    candidates = {row["candidate_id"]: row for row in inputs["candidate_inputs"]}
    bindings = {row["candidate_id"]: row for row in vectors["vector_bindings"]}
    require(
        len(candidates) == len(bindings) == 33 and candidates.keys() == bindings.keys(),
        "C188_CANDIDATE_SET_MISMATCH",
    )
    cache = vectors["cache_files"]
    metadata = json.loads(read_bound(
        cache["metadata"]["path"], cache["metadata"]["file_bytes_sha256"]
    ))
    array_raw = read_bound(cache["array"]["path"], cache["array"]["file_bytes_sha256"])
    array = np.load(root / cache["array"]["path"], allow_pickle=False)
    require(
        sha((root / cache["array"]["path"]).read_bytes()) == sha(array_raw)
        and array.shape == (1149, 384) and array.dtype == np.float32
        and metadata["cache_key"] == vectors["cache_key"],
        "C188_VECTOR_CACHE_MISMATCH",
    )
    del array_raw

    nb_path = "notebooks/phase2/16_hybrid_retrieval.ipynb"
    notebook = json.loads(read_bound(nb_path, nb_sha))
    scope = dict(globals())
    scope.pop("qdrant_client", None)
    scope["print"] = lambda *args, **kwargs: None

    def parse_cell(index):
        cell = notebook["cells"][index]
        require(cell["cell_type"] == "code", f"C188_EXPECTED_CODE_CELL:{index}")
        return ast.parse("".join(cell["source"]), filename=f"{nb_path}:{index}")

    def execute_nodes(index, nodes):
        exec(compile(
            ast.Module(body=nodes, type_ignores=[]),
            f"{nb_path}:{index}", "exec", optimize=0,
        ), scope)

    outcomes = []
    attempts = 0
    setup_blocker = None
    close_error = None
    stage = "LOAD_PINNED_NB16_SETUP"

    try:
        for index in (3, 5, 7, 9, 11, 13):
            stage = f"NB16_SETUP_CELL_{index}"
            execute_nodes(index, parse_cell(index).body)
        require(
            Path(scope["PROJECT_ROOT"]).resolve() == root,
            "C188_NB16_PROJECT_ROOT_MISMATCH",
        )

        stage = "OPEN_PINNED_DENSE_INDEX"
        execute_nodes(15, parse_cell(15).body)
        dense_path = Path(scope["dense_storage_path"]).resolve()
        require(
            dense_path.is_relative_to(root) and dense_path.is_dir(),
            "C188_DENSE_STORAGE_MISMATCH",
        )

        names = {
            "frozen_bm25_scores", "frozen_sparse_retrieve", "frozen_dense_retrieve",
            "frozen_minmax_scores", "frozen_weighted_fusion",
        }
        definitions = [
            node for node in parse_cell(37).body
            if isinstance(node, ast.FunctionDef) and node.name in names
        ]
        require(
            len(definitions) == len(names)
            and {node.name for node in definitions} == names,
            "C188_FROZEN_FUNCTIONS_MISSING",
        )
        scope.update(
            FROZEN_CANDIDATE_DEPTH=100,
            FROZEN_DENSE_WEIGHT=0.4,
            FROZEN_SPARSE_WEIGHT=0.6,
        )
        execute_nodes(37, definitions)

        stage = "RUN_BATCH"
        for candidate in sorted(candidates):
            row = candidates[candidate]
            binding = bindings[candidate]
            outcome = {"candidate_id": candidate}
            try:
                question = row["question"].strip()
                index = binding["cache_row_index_zero_based"]
                require(
                    row["split"] == "test"
                    and metadata["candidate_ids"][index] == candidate
                    and sha(question.encode("utf-8"))
                    == binding["embedded_question_sha256"]
                    == row["ranker_question_sha256"]
                    and scope["golden_record_by_candidate_id"][candidate][
                        "evaluation_content"
                    ]["question"] == row["question"],
                    "C188_QUERY_BINDING_MISMATCH",
                )
                vector = array[index].copy()
                require(
                    sha(vector.tobytes()) == binding["query_vector_sha256"]
                    and bool(np.isfinite(vector).all())
                    and bool(np.isclose(np.linalg.norm(vector), 1.0)),
                    "C188_VECTOR_BINDING_MISMATCH",
                )

                sparse = scope["frozen_sparse_retrieve"](question, 100)
                attempts += 1
                dense = scope["frozen_dense_retrieve"](vector, 100)
                fused = scope["frozen_weighted_fusion"](
                    dense["ranked_chunk_ids"], dense["ranked_scores"],
                    sparse["ranked_chunk_ids"], sparse["ranked_scores"],
                )
                observed = fused["ranked_chunk_ids"][:100]
                require(
                    len(observed) == len(set(observed)) == 100,
                    "C188_OBSERVED_DEPTH_OR_DUPLICATE_MISMATCH",
                )
                expected = row["hybrid100_order"]
                differences = [
                    {"rank": rank, "expected": old, "observed": new}
                    for rank, (old, new) in enumerate(zip(expected, observed), 1)
                    if old != new
                ]
                pool_match = set(observed[:50]) == set(row["pool50_order"])
                matched = observed == expected
                outcome.update(
                    status="FULL_ORDER_MATCHED" if matched else "FULL_ORDER_DIFFERS",
                    question_sha256=row["question_sha256"],
                    query_vector_sha256=binding["query_vector_sha256"],
                    observed_hybrid100_order=observed,
                    full_order_matches=matched,
                    pool50_membership_matches=pool_match,
                    different_position_count=len(differences),
                    first_difference=differences[0] if differences else None,
                )
            except (OSError, ValueError, KeyError, TypeError, RuntimeError) as error:
                outcome.update(
                    status="QUERY_BLOCKED",
                    blocker=f"{type(error).__name__}:{error}",
                )
            outcomes.append(outcome)

    except Exception as error:
        setup_blocker = {
            "stage": stage, "type": type(error).__name__, "detail": str(error),
        }
    finally:
        client = scope.get("qdrant_client")
        if client is not None:
            try:
                client.close()
            except Exception as error:
                close_error = type(error).__name__

    matched = sum(row["status"] == "FULL_ORDER_MATCHED" for row in outcomes)
    all_matched = (
        setup_blocker is None and close_error is None
        and len(outcomes) == matched == 33
        and all(row.get("pool50_membership_matches") is True for row in outcomes)
    )
    result = {
        "cell": "C188",
        "roadmap_subtask": "R11.3",
        "checklist_item": 12,
        "status": "ALL_HYBRID100_ORDERS_MATCHED" if all_matched else "HYBRID100_BATCH_REQUIRES_REVIEW",
        "input_file_hashes": {"C186": c186_sha, "C187": c187_sha, "NB16": nb_sha},
        "candidate_count": 33,
        "checked_candidate_count": len(outcomes),
        "full_order_match_count": matched,
        "setup_blocker": setup_blocker,
        "qdrant_close_error": close_error,
        "qdrant_queries_attempted": attempts,
        "embedding_calls": 0,
        "new_api_calls": 0,
        "new_response_scenario_results": 0,
        "candidate_results": outcomes,
        "done_item_count": 9,
        "open_item_count": 3,
        "open_item_ids": [9, 10, 12],
        "selected_version_gate": "BLOCKED",
        "next_dependency": (
            "BIND_LOCAL100_QUOTA100_AND_CHAR_PROVIDERS_FOR_THE_BATCH"
            if all_matched else "RESOLVE_RECORDED_HYBRID_ORDER_OR_RUNTIME_DIFFERENCES"
        ),
    }
    raw = (
        json.dumps(result, ensure_ascii=False, sort_keys=True, indent=2) + "\n"
    ).encode("utf-8")
    digest = sha(raw)
    relative = reports / f"C188.R11_remaining_hybrid100_comparison.{digest}.json"
    path = root / relative
    try:
        with path.open("xb") as stream:
            stream.write(raw)
    except FileExistsError:
        pass
    require(path.read_bytes() == raw, "C188_OUTPUT_RELOAD_MISMATCH")
    result["output"] = {
        "path": relative.as_posix(),
        "file_bytes_sha256": digest,
        "reload_validation": "PASS",
    }
    summary = {key: value for key, value in result.items() if key != "candidate_results"}
    summary["candidate_issues"] = [
        {key: value for key, value in row.items() if key != "observed_hybrid100_order"}
        for row in outcomes if row["status"] != "FULL_ORDER_MATCHED"
    ]
    print(json.dumps(summary, ensure_ascii=False, indent=2))
    return result


V3_C188 = run_v3_c188()
````

````python
# NB22E-C189 — Compute the remaining selected ranking inputs.
# Reuse pool50 scores and checkpoint complete candidate results.

def run_v3_c189():
    def require(condition, code):
        if not condition:
            raise ValueError(code)

    def encode(value):
        return json.dumps(
            value,
            ensure_ascii=False,
            sort_keys=True,
            separators=(",", ":"),
            allow_nan=False,
        ).encode("utf-8")

    def sha(raw):
        return hashlib.sha256(raw).hexdigest()

    reports = Path("reports/phase2/quality_improvement/NB22E.lab.v3")
    cwd = Path.cwd()
    root = next(
        (
            path.resolve()
            for path in (cwd, *cwd.parents)
            if (path / reports).is_dir()
        ),
        None,
    )
    require(root is not None, "C189_PROJECT_ROOT_NOT_FOUND")

    def read_bound(relative, digest):
        path = (root / relative).resolve()
        require(path.is_relative_to(root), "C189_PATH_OUTSIDE_PROJECT")
        raw = path.read_bytes()
        require(sha(raw) == digest, f"C189_HASH_MISMATCH:{relative}")
        return raw

    def read_report(stem, digest):
        relative = reports / f"{stem}.{digest}.json"
        return json.loads(read_bound(relative, digest))

    def atomic_create(path, raw):
        path.parent.mkdir(parents=True, exist_ok=True)
        descriptor, temporary = tempfile.mkstemp(
            prefix=".C189-",
            dir=str(path.parent),
        )
        try:
            with os.fdopen(descriptor, "wb") as stream:
                stream.write(raw)
                stream.flush()
                os.fsync(stream.fileno())
            try:
                os.link(temporary, path)
            except FileExistsError:
                pass
            require(path.read_bytes() == raw, "C189_EXISTING_OUTPUT_DIFFERS")
        finally:
            Path(temporary).unlink(missing_ok=True)

    c186_sha = "5cb0fa8851d28b1fa459f01cfbf6355fc32f1db9d5f484a133685b93d6315821"
    c188_sha = "f24429fa783c39db82101f46f3078e8d1b3294b04973ebdc6e52abd7956264f8"
    c118_sha = "21aee5d3199de8d81afc83b58f96ff67df21fc47b019132c52b8d67235524c4d"
    c119_sha = "967178107d6a4fab03ecb5ffd74f79ceebfd716dc265d9726339164d06a3647b"

    inputs = read_report(
        "C186.R11_remaining_scenario_input_preflight", c186_sha
    )
    hybrid = read_report(
        "C188.R11_remaining_hybrid100_comparison", c188_sha
    )
    config = read_report(
        "C118.R11_local100_model_preflight", c118_sha
    )["expected_runtime"]
    parity = read_report(
        "C119.R11_local100_four_pair_parity", c119_sha
    )

    require(
        hybrid["status"] == "ALL_HYBRID100_ORDERS_MATCHED"
        and hybrid["input_file_hashes"]["C186"] == c186_sha
        and hybrid["full_order_match_count"] == inputs["candidate_count"] == 33
        and parity["score_parity_verified"] is True
        and parity["input_file_hashes"]["C118"] == c118_sha
        and config["batch_size"] == 4
        and config["maximum_sequence_length"] == 512,
        "C189_INPUT_OR_RUNTIME_CONTRACT_MISMATCH",
    )

    corpus_path = (
        "data/phase2/chunks/nb11_benchmark_v1/"
        "baseline_phase1_char_safe_v1.jsonl"
    )
    corpus_sha = inputs["input_file_hashes"]["chunks"]
    chunks = {}
    for line in read_bound(corpus_path, corpus_sha).splitlines():
        if line.strip():
            row = json.loads(line)
            chunk_id = row["chunk_id"]
            require(chunk_id not in chunks, "C189_DUPLICATE_CHUNK")
            chunks[chunk_id] = row
    require(len(chunks) == 3488, "C189_CORPUS_COUNT_MISMATCH")

    def load_functions(path, digest, index, names, configuration):
        notebook = json.loads(read_bound(path, digest))
        source = "".join(notebook["cells"][index]["source"])
        nodes = [
            node
            for node in ast.parse(source).body
            if isinstance(node, ast.FunctionDef) and node.name in names
        ]
        require(
            len(nodes) == len(names)
            and {node.name for node in nodes} == names,
            f"C189_FROZEN_FUNCTIONS_MISSING:{path}",
        )
        scope = dict(globals())
        scope.update(require=require, CONFIG=configuration)
        module = ast.Module(body=nodes, type_ignores=[])
        exec(compile(module, path, "exec"), scope)
        return scope

    c_sha = "e658b973c057d945add9a9f7c96c6a98b36aae1b2fca70cf958925df4de7315c"
    d_sha = "fff5bd072a4977820882045451b61d66f8621ef9a4918775e4534800f0ae1480"
    quota_spec = {
        "family": "quota100",
        "k": 10,
        "weight": 0.75,
        "newcomer_limit": 1,
    }
    char_config = {
        "char_max_df_fraction": 0.30,
        "bm25": {"k1": 1.2, "b": 0.3},
        "rrf_k": 10,
        "incumbent_rank_weight": 0.75,
    }

    quota_scope = load_functions(
        "notebooks/phase2/22c_retrieval_diagnosis_and_improvement.ipynb",
        c_sha,
        6,
        {"derived_order"},
        {},
    )
    char_scope = load_functions(
        "notebooks/phase2/22d_full_corpus_retrieval_expansion.ipynb",
        d_sha,
        12,
        {
            "normalized_words",
            "char_tokens",
            "build_index",
            "lexical_search",
            "lexical_fusion",
        },
        char_config,
    )
    char_index = char_scope["build_index"](chunks, "char")

    runtime = globals().get("ort")
    if runtime is None:
        runtime = globals().get("onnxruntime")
    require(
        runtime is not None
        and runtime.__version__ == config["onnxruntime_version"],
        "C189_ONNXRUNTIME_VERSION_MISMATCH",
    )

    hub_cache = os.environ.get("HF_HUB_CACHE")
    hf_home = os.environ.get("HF_HOME")
    if hub_cache:
        cache_root = Path(hub_cache)
    elif hf_home:
        cache_root = Path(hf_home) / "hub"
    else:
        cache_root = Path.home() / ".cache" / "huggingface" / "hub"

    model_folder = "models--" + config["model_id"].replace("/", "--")
    snapshot = cache_root / model_folder / "snapshots" / config["model_revision"]
    model_path = snapshot / "onnx" / "model_quint8_avx2.onnx"
    require(
        sha(model_path.read_bytes()) == config["onnx_sha256"],
        "C189_MODEL_HASH_MISMATCH",
    )

    contract = {
        "C186": c186_sha,
        "C188": c188_sha,
        "C118": c118_sha,
        "C119": c119_sha,
        "NB22C": c_sha,
        "NB22D": d_sha,
        "chunks": corpus_sha,
        "local_runtime": config,
        "quota": quota_spec,
        "char": char_config,
        "execution_profile": "CPU_4_THREADS_BATCH4_MAX512",
    }
    contract_sha = sha(encode(contract))
    folder = root / reports / "C189.R11_rank_checkpoints" / contract_sha
    matched = {
        row["candidate_id"]: row
        for row in hybrid["candidate_results"]
    }
    require(len(matched) == 33, "C189_HYBRID_RESULT_POPULATION_MISMATCH")

    def compute_orders(hybrid_ids, pool_ids, scores, question):
        local = sorted(hybrid_ids, key=lambda cid: (-scores[cid], cid))
        quota = quota_scope["derived_order"](
            hybrid_ids, local, pool_ids, quota_spec
        )
        channel = char_scope["lexical_search"](
            char_index, question, "char"
        )
        fused = char_scope["lexical_fusion"](
            quota, {"char": channel}, {"channels": ["char"]}
        )
        require(
            len(quota) == len(set(quota)) == 100
            and set(quota) == set(hybrid_ids)
            and len(fused[:10]) == len(set(fused[:10])) == 10
            and set(fused) == set(quota) | set(channel),
            "C189_FINAL_ORDER_CONTRACT_MISMATCH",
        )
        return {
            "local100_order": local,
            "quota100_order": quota,
            "char_channel_order": channel,
            "char_rrf_order": fused,
            "char_rrf_top10": fused[:10],
            "char_rrf_top10_text_sha256": [
                sha(chunks[cid]["content"].encode("utf-8"))
                for cid in fused[:10]
            ],
        }

    session = None
    tokenizer = None
    names = None
    receipts = []
    blockers = []
    new_pairs = 0
    inference_calls = 0
    reused = 0

    for item in inputs["candidate_inputs"]:
        candidate = item["candidate_id"]
        try:
            question = item["question"].strip()
            hybrid_ids = item["hybrid100_order"]
            pool_ids = item["pool50_order"]
            pool_scores = item["pool50_scores_by_chunk_id"]
            require(
                item["status"] == "HISTORICAL_INPUTS_BOUND"
                and sha(question.encode("utf-8")) == item["ranker_question_sha256"]
                and matched[candidate]["status"] == "FULL_ORDER_MATCHED"
                and matched[candidate]["observed_hybrid100_order"] == hybrid_ids
                and set(pool_ids) == set(hybrid_ids[:50]) == set(pool_scores)
                and all(math.isfinite(value) for value in pool_scores.values())
                and [
                    sha(chunks[cid]["content"].encode("utf-8"))
                    for cid in hybrid_ids
                ] == item["hybrid100_text_sha256"],
                "C189_CANDIDATE_INPUT_MISMATCH",
            )

            input_sha = sha(encode(item))
            checkpoint_path = folder / f"{input_sha}.json"

            if checkpoint_path.exists():
                raw = checkpoint_path.read_bytes()
                saved = json.loads(raw)
                scores = saved["local100_scores_by_chunk_id"]
                require(
                    saved["contract_sha256"] == contract_sha
                    and saved["candidate_input_sha256"] == input_sha
                    and saved["candidate_id"] == candidate
                    and saved["question_sha256"] == item["question_sha256"]
                    and saved["ranker_question_sha256"] == item["ranker_question_sha256"]
                    and saved["hybrid100_order"] == hybrid_ids
                    and saved["pool50_order"] == pool_ids
                    and set(scores) == set(hybrid_ids)
                    and all(math.isfinite(value) for value in scores.values())
                    and all(
                        scores[cid] == value
                        for cid, value in pool_scores.items()
                    ),
                    "C189_CHECKPOINT_BINDING_MISMATCH",
                )
                orders = compute_orders(
                    hybrid_ids, pool_ids, scores, question
                )
                require(
                    all(saved[key] == value for key, value in orders.items()),
                    "C189_CHECKPOINT_ORDER_MISMATCH",
                )
                reused += 1
            else:
                if session is None:
                    tokenizer = AutoTokenizer.from_pretrained(
                        str(snapshot),
                        local_files_only=True,
                        trust_remote_code=False,
                    )
                    options = runtime.SessionOptions()
                    options.intra_op_num_threads = 4
                    options.inter_op_num_threads = 1
                    options.execution_mode = runtime.ExecutionMode.ORT_SEQUENTIAL
                    options.graph_optimization_level = (
                        runtime.GraphOptimizationLevel.ORT_ENABLE_ALL
                    )
                    session = runtime.InferenceSession(
                        str(model_path),
                        sess_options=options,
                        providers=["CPUExecutionProvider"],
                    )
                    names = {entry.name for entry in session.get_inputs()}
                    require(
                        {"input_ids", "attention_mask"} <= names,
                        "C189_ONNX_INPUT_MISMATCH",
                    )

                scores = dict(pool_scores)
                missing = [
                    cid for cid in hybrid_ids if cid not in scores
                ]
                require(
                    len(missing) == 50,
                    "C189_MISSING_PAIR_COUNT_MISMATCH",
                )

                for start in range(0, len(missing), 4):
                    batch = missing[start:start + 4]
                    encoded = tokenizer(
                        [question] * len(batch),
                        [chunks[cid]["content"] for cid in batch],
                        padding=True,
                        truncation=True,
                        max_length=512,
                        return_tensors="np",
                    )
                    require(
                        names <= set(encoded),
                        "C189_TOKENIZER_INPUT_MISMATCH",
                    )
                    feed = {
                        name: np.asarray(encoded[name], dtype=np.int64)
                        for name in names
                    }
                    inference_calls += 1
                    values = np.asarray(
                        session.run(None, feed)[0],
                        dtype=np.float64,
                    ).reshape(-1)
                    require(
                        values.shape == (len(batch),)
                        and bool(np.isfinite(values).all()),
                        "C189_INVALID_LOCAL_SCORES",
                    )
                    scores.update({
                        cid: float(value)
                        for cid, value in zip(batch, values)
                    })
                    new_pairs += len(batch)

                orders = compute_orders(
                    hybrid_ids, pool_ids, scores, question
                )
                saved = {
                    "candidate_id": candidate,
                    "question_sha256": item["question_sha256"],
                    "ranker_question_sha256": item["ranker_question_sha256"],
                    "contract_sha256": contract_sha,
                    "candidate_input_sha256": input_sha,
                    "hybrid100_order": hybrid_ids,
                    "pool50_order": pool_ids,
                    "local100_scores_by_chunk_id": scores,
                    **orders,
                }
                raw = encode(saved)
                atomic_create(checkpoint_path, raw)

            receipts.append({
                "candidate_id": candidate,
                "path": checkpoint_path.relative_to(root).as_posix(),
                "file_bytes_sha256": sha(raw),
                "char_channel_depth": len(saved["char_channel_order"]),
                "char_rrf_depth": len(saved["char_rrf_order"]),
            })
            print(
                f"C189 {len(receipts)}/33 complete | "
                f"new_pairs={new_pairs} | reused={reused}",
                flush=True,
            )

        except Exception as error:
            blockers.append({
                "candidate_id": candidate,
                "type": type(error).__name__,
                "detail": str(error),
            })
            print(f"C189 blocked: {candidate} | {type(error).__name__}", flush=True)

    complete = len(receipts) == 33 and not blockers
    result = {
        "cell": "C189",
        "roadmap_subtask": "R11.3",
        "checklist_item": 12,
        "status": (
            "ALL_SELECTED_RANK_INPUTS_BUILT"
            if complete else "RANK_INPUT_BATCH_PARTIAL"
        ),
        "contract": contract,
        "contract_sha256": contract_sha,
        "candidate_count": 33,
        "completed_candidate_count": len(receipts),
        "reused_candidate_checkpoints": reused,
        "new_local_pairs_scored": new_pairs,
        "model_inference_calls_attempted": inference_calls,
        "candidate_receipts": receipts,
        "blocked_candidates": blockers,
        "qdrant_queries_attempted": 0,
        "new_api_calls": 0,
        "new_response_scenario_results": 0,
        "historical_char_rrf_order_equivalence_claimed": False,
        "done_item_count": 9,
        "open_item_count": 3,
        "open_item_ids": [9, 10, 12],
        "selected_version_gate": "BLOCKED",
        "next_dependency": (
            "EXERCISE_BATCH_PRODUCT_INPUT_BOUNDARIES_BEFORE_LIVE_GENERATION"
            if complete else "RESOLVE_RECORDED_CANDIDATE_BLOCKERS"
        ),
    }

    raw = encode(result)
    digest = sha(raw)
    relative = reports / f"C189.R11_remaining_selected_rank_inputs.{digest}.json"
    atomic_create(root / relative, raw)
    result["output"] = {
        "path": relative.as_posix(),
        "file_bytes_sha256": digest,
        "reload_validation": "PASS",
    }

    summary = {
        key: value
        for key, value in result.items()
        if key not in {"candidate_receipts", "contract"}
    }
    print(json.dumps(summary, ensure_ascii=False, indent=2))
    return result


V3_C189 = run_v3_c189()
````

````python
# NB22E-C190 — Verify all stored ranking inputs at the product generation boundary.
# Persist and reread each input receipt while capturing the exact rendered context.

def run_v3_c190():
    def require(ok, code):
        if not ok:
            raise ValueError(code)

    def encode(value):
        return json.dumps(
            value, ensure_ascii=False, sort_keys=True,
            separators=(",", ":"), allow_nan=False,
        ).encode("utf-8")

    def sha(raw):
        return hashlib.sha256(raw).hexdigest()

    def text_sha(text):
        return sha(text.encode("utf-8"))

    reports = Path("reports/phase2/quality_improvement/NB22E.lab.v3")
    cwd = Path.cwd()
    root = next(
        (p.resolve() for p in (cwd, *cwd.parents) if (p / reports).is_dir()),
        None,
    )
    require(root is not None, "C190_PROJECT_ROOT_NOT_FOUND")

    def read_bound(relative, digest):
        path = (root / relative).resolve()
        require(path.is_relative_to(root), "C190_PATH_OUTSIDE_PROJECT")
        raw = path.read_bytes()
        require(sha(raw) == digest, f"C190_HASH_MISMATCH:{relative}")
        return raw

    def load_ref(pin):
        return json.loads(read_bound(pin["path"], pin["file_bytes_sha256"]))

    def load_report(stem, digest):
        return json.loads(read_bound(
            reports / f"{stem}.{digest}.json", digest,
        ))

    def save(stem, value):
        raw = encode(value)
        digest = sha(raw)
        relative = reports / f"{stem}.{digest}.json"
        path = root / relative
        descriptor, temporary = tempfile.mkstemp(
            prefix=".C190-", dir=str(path.parent),
        )
        try:
            with os.fdopen(descriptor, "wb") as stream:
                stream.write(raw)
                stream.flush()
                os.fsync(stream.fileno())
            try:
                os.link(temporary, path)
            except FileExistsError:
                pass
            require(path.read_bytes() == raw, "C190_OUTPUT_RELOAD_MISMATCH")
        finally:
            Path(temporary).unlink(missing_ok=True)
        return {
            "path": relative.as_posix(),
            "file_bytes_sha256": digest,
            "reload_validation": "PASS",
        }

    c186_sha = "5cb0fa8851d28b1fa459f01cfbf6355fc32f1db9d5f484a133685b93d6315821"
    c188_sha = "f24429fa783c39db82101f46f3078e8d1b3294b04973ebdc6e52abd7956264f8"
    c189_sha = "54b17688cd7f96885984b8f6a9df2f64b53bd636090128269ea12b54c0422f5f"
    c139_sha = "3b5a43d5476dc28de79eed5fc35adc6e2a6834bbd9c6a341c8efa93b18669634"

    inputs = load_report("C186.R11_remaining_scenario_input_preflight", c186_sha)
    hybrid = load_report("C188.R11_remaining_hybrid100_comparison", c188_sha)
    ranks = load_report("C189.R11_remaining_selected_rank_inputs", c189_sha)
    route = load_report("C139.R11_persisted_opt_in_caller", c139_sha)
    require(
        ranks["status"] == "ALL_SELECTED_RANK_INPUTS_BUILT"
        and ranks["completed_candidate_count"] == 33
        and ranks["contract"]["C186"] == c186_sha
        and ranks["contract"]["C188"] == c188_sha
        and not ranks["blocked_candidates"],
        "C190_SUCCESSFUL_C189_REQUIRED",
    )

    pipeline_relative = "src/pipeline/rag_pipeline.py"
    pipeline_sha = "99ee048d00a7ad1884540fa297eb41df86ad4e54f6c49f26385ebb6b2ed504f6"
    pipeline_raw = read_bound(pipeline_relative, pipeline_sha)
    if str(root) not in sys.path:
        sys.path.insert(0, str(root))
    importlib.invalidate_caches()
    module = importlib.import_module("src.pipeline.rag_pipeline")
    require(
        Path(module.__file__).resolve() == root / pipeline_relative,
        "C190_IMPORTED_PRODUCT_PATH_MISMATCH",
    )
    pipeline_class = module.RAGPipeline

    # Verify that the loaded methods correspond to the pinned source.
    code_type = type(run_v3_c190.__code__)
    compiled = compile(
        pipeline_raw, pipeline_relative, "exec", dont_inherit=True,
    )
    class_code = next(
        value for value in compiled.co_consts
        if isinstance(value, code_type) and value.co_name == "RAGPipeline"
    )
    method_codes = {
        value.co_name: value for value in class_code.co_consts
        if isinstance(value, code_type)
    }

    def signature(code):
        return (
            code.co_code,
            tuple(
                signature(value) if isinstance(value, code_type) else value
                for value in code.co_consts
            ),
            code.co_names, code.co_varnames, code.co_freevars, code.co_cellvars,
            code.co_flags, code.co_argcount, code.co_posonlyargcount,
            code.co_kwonlyargcount, code.co_exceptiontable,
        )

    for name in ("__init__", "ask_with_char_rrf", "build_context", "close"):
        require(
            signature(getattr(pipeline_class, name).__code__)
            == signature(method_codes[name]),
            f"C190_LOADED_METHOD_MISMATCH:{name}",
        )

    route_pin = route["module"]
    namespace = {}
    exec(compile(
        read_bound(route_pin["path"], route_pin["file_bytes_sha256"]),
        route_pin["path"], "exec", dont_inherit=True,
    ), namespace)
    caller = namespace[route_pin["export"]]

    c148_sha = "8352cfa49bedc99c50335f57a66051a3613bdb1cf3618200dff771a8c1d654f4"
    dense_reference = load_report("C148.R11_target_test_hybrid100_probe", c148_sha)
    storage = (root / dense_reference["dense_storage_path"]).resolve()
    require(storage.is_relative_to(root) and storage.is_dir(), "C190_DENSE_STORAGE_MISSING")

    corpus_path = "data/phase2/chunks/nb11_benchmark_v1/baseline_phase1_char_safe_v1.jsonl"
    corpus = {}
    for line in read_bound(
        corpus_path, inputs["input_file_hashes"]["chunks"]
    ).splitlines():
        if line.strip():
            row = json.loads(line)
            require(row["chunk_id"] not in corpus, "C190_DUPLICATE_CHUNK")
            corpus[row["chunk_id"]] = row

    items = {row["candidate_id"]: row for row in inputs["candidate_inputs"]}
    rank_pins = {row["candidate_id"]: row for row in ranks["candidate_receipts"]}
    hybrid_rows = {row["candidate_id"]: row for row in hybrid["candidate_results"]}
    require(
        len(items) == len(rank_pins) == len(hybrid_rows) == 33
        and items.keys() == rank_pins.keys() == hybrid_rows.keys(),
        "C190_CANDIDATE_SET_MISMATCH",
    )

    class StopAtGeneration(Exception):
        pass

    completed = []
    blockers = []
    pipeline = pipeline_class(
        vectorstore_path=storage,
        collection_name=dense_reference["dense_collection"],
        use_cache=False,
    )
    try:
        for candidate in sorted(items):
            events = []
            input_ref = None
            try:
                item = items[candidate]
                rank_pin = rank_pins[candidate]
                rank = load_ref(rank_pin)
                question = item["question"]
                question_hash = item["question_sha256"]
                hybrid_ids = item["hybrid100_order"]
                pool_ids = item["pool50_order"]
                pool_scores = item["pool50_scores_by_chunk_id"]
                top10 = rank["char_rrf_top10"]
                require(
                    rank["candidate_id"] == candidate
                    and rank["contract_sha256"] == ranks["contract_sha256"]
                    and rank["candidate_input_sha256"] == sha(encode(item))
                    and rank["hybrid100_order"] == hybrid_ids
                    and rank["pool50_order"] == pool_ids
                    and hybrid_rows[candidate]["observed_hybrid100_order"] == hybrid_ids
                    and top10 == rank["char_rrf_order"][:10]
                    and len(top10) == len(set(top10)) == 10
                    and text_sha(question) == question_hash,
                    "C190_CANDIDATE_RANK_BINDING_MISMATCH",
                )
                chunks = [
                    {
                        "chunk_id": cid,
                        "text": corpus[cid]["content"],
                        "title": corpus[cid].get("title"),
                        "url": corpus[cid].get("url"),
                    }
                    for cid in top10
                ]
                text_hashes = [text_sha(chunk["text"]) for chunk in chunks]
                require(
                    text_hashes == rank["char_rrf_top10_text_sha256"],
                    "C190_CHUNK_TEXT_MISMATCH",
                )

                def query_check(question, candidate_id):
                    require(
                        candidate_id == candidate and text_sha(question) == question_hash,
                        "C190_PROVIDER_QUERY_MISMATCH",
                    )

                def hybrid_provider(*, question, candidate_id):
                    query_check(question, candidate_id)
                    events.append("bound_hybrid100_supplied")
                    return {
                        "candidate_id": candidate,
                        "question_sha256": question_hash,
                        "ordered_chunk_ids": hybrid_ids,
                        "receipt_sha256": c188_sha,
                    }

                pool_pin = save("C190.R11_bound_pool50", {
                    "candidate_id": candidate,
                    "question_sha256": question_hash,
                    "source_row": item["source_rows"]["reranker"],
                    "hybrid100_order": hybrid_ids,
                    "ordered_chunk_ids": pool_ids,
                    "scores_by_chunk_id": pool_scores,
                })

                def pool_provider(*, question, candidate_id, hybrid100):
                    query_check(question, candidate_id)
                    require(list(hybrid100) == hybrid_ids, "C190_HYBRID_INPUT_MISMATCH")
                    events.append("bound_pool50_supplied")
                    return {
                        "candidate_id": candidate,
                        "question_sha256": question_hash,
                        "ordered_chunk_ids": pool_ids,
                        "scores_by_chunk_id": pool_scores,
                        "receipt_sha256": pool_pin["file_bytes_sha256"],
                    }

                runtime_identity = {
                    "scope": "BOUND_RANKS_STOP_BEFORE_GENERATION",
                    "pipeline_source_sha256": pipeline_sha,
                    "route_source_sha256": route_pin["file_bytes_sha256"],
                    "rank_receipt_sha256": rank_pin["file_bytes_sha256"],
                    "rank_contract_sha256": ranks["contract_sha256"],
                }

                def rank_provider(
                    *, question, candidate_id, top_k, hybrid100,
                    pool50_order, pool50_scores
                ):
                    query_check(question, candidate_id)
                    require(
                        top_k == 10 and list(hybrid100) == hybrid_ids
                        and list(pool50_order) == pool_ids
                        and pool50_scores == pool_scores,
                        "C190_RANK_PROVIDER_INPUT_MISMATCH",
                    )
                    events.append("bound_char_rrf_supplied")
                    return {
                        "candidate_id": candidate,
                        "question_sha256": question_hash,
                        "selected_variant": "char_rrf",
                        "runtime_identity": dict(runtime_identity),
                        "ordered_chunks": chunks,
                    }

                def write_receipt(role, payload):
                    nonlocal input_ref
                    require(
                        role == "input" and input_ref is None,
                        "C190_UNEXPECTED_RECEIPT_WRITE",
                    )
                    input_ref = save("C190.R11_product_input_receipt", payload)
                    events.append("input_receipt_written")
                    return input_ref["file_bytes_sha256"]

                captured = {}

                class Boundary:
                    def generate_answer(self, *, question, context):
                        query_check(question, candidate)
                        require(input_ref is not None, "C190_INPUT_RECEIPT_REQUIRED")
                        recorded = load_ref(input_ref)
                        require(
                            recorded["candidate_id"] == candidate
                            and recorded["question_sha256"] == question_hash
                            and recorded["selected_variant"] == "char_rrf"
                            and recorded["ordered_chunk_ids"] == top10
                            and recorded["ordered_chunk_text_sha256"] == text_hashes
                            and recorded["rendered_context_sha256"] == text_sha(context)
                            and recorded["runtime_identity"] == {
                                **runtime_identity,
                                "hybrid100_provider_receipt_sha256": c188_sha,
                                "pool50_provider_receipt_sha256": pool_pin["file_bytes_sha256"],
                            },
                            "C190_GENERATION_BOUNDARY_MISMATCH",
                        )
                        captured["context"] = context
                        events.append("generation_boundary_verified")
                        raise StopAtGeneration()

                pipeline.generator = Boundary()
                try:
                    caller(
                        pipeline,
                        question=question,
                        candidate_id=candidate,
                        hybrid100_provider=hybrid_provider,
                        pool50_provider=pool_provider,
                        rank_provider=rank_provider,
                        write_receipt=write_receipt,
                        top_k=10,
                    )
                except StopAtGeneration:
                    pass
                else:
                    raise ValueError("C190_EXPECTED_GENERATION_BOUNDARY_STOP")

                require(events == [
                    "bound_hybrid100_supplied",
                    "bound_pool50_supplied",
                    "bound_char_rrf_supplied",
                    "input_receipt_written",
                    "generation_boundary_verified",
                ], "C190_EVENT_ORDER_MISMATCH")

                prepared = save("C190.R11_prepared_product_context", {
                    "candidate_id": candidate,
                    "question": question,
                    "question_sha256": question_hash,
                    "input_receipt": input_ref,
                    "rank_receipt": rank_pin,
                    "rendered_context": captured["context"],
                    "rendered_context_sha256": text_sha(captured["context"]),
                    "scenario_specs": [
                        spec for spec in inputs["scenario_specs"]
                        if spec["candidate_id"] == candidate
                    ],
                    "events": events,
                    "responses_generated": 0,
                })
                completed.append({"candidate_id": candidate, **prepared})

            except Exception as error:
                blockers.append({
                    "candidate_id": candidate,
                    "type": type(error).__name__,
                    "detail": str(error),
                })
    finally:
        pipeline.close()

    result = {
        "cell": "C190",
        "roadmap_subtask": "R11.3",
        "checklist_item": 12,
        "status": (
            "ALL_PRODUCT_INPUT_BOUNDARIES_VERIFIED"
            if len(completed) == 33 and not blockers
            else "PRODUCT_INPUT_BOUNDARY_BATCH_PARTIAL"
        ),
        "input_file_hashes": {
            "C186": c186_sha,
            "C188": c188_sha,
            "C189": c189_sha,
            "C139": c139_sha,
            "pipeline_source": pipeline_sha,
        },
        "candidate_count": 33,
        "verified_candidate_count": len(completed),
        "prepared_context_receipts": completed,
        "blocked_candidates": blockers,
        "rank_input_mode": "VERSION_BOUND_STORED_C189_RANKS",
        "responses_generated": 0,
        "qdrant_queries_attempted": 0,
        "new_local_pairs_scored": 0,
        "new_api_calls": 0,
        "new_response_scenario_results": 0,
        "done_item_count": 9,
        "open_item_count": 3,
        "open_item_ids": [9, 10, 12],
        "selected_version_gate": "BLOCKED",
    }
    result["output"] = save("C190.R11_batch_product_input_boundaries", result)
    print(json.dumps(
        {
            key: value for key, value in result.items()
            if key != "prepared_context_receipts"
        },
        ensure_ascii=False, indent=2,
    ))
    return result


V3_C190 = run_v3_c190()
````

### NB22E-C191 — آماده‌سازی بستهٔ درخواست‌های DeepSeek و برآورد هزینه

برآورد هزینهٔ هر درخواست و کل بسته با نرخ تاریخی C042 ثبت می‌شود.
سقف پیشنهادی کل بسته ۱۰۰٬۰۰۰ تومان است. ارسال بعدی باید رسید ورودی

این سل آفلاین است و نتیجهٔ تازه‌ای برای سناریوها ثبت نمی‌کند.

````python
# NB22E-C191 — Prepare the remaining DeepSeek requests and batch cost estimate.
# Bind each request to its verified product context without dispatching it.

def run_v3_c191():
    def require(ok, code):
        if not ok:
            raise ValueError(code)

    def encode(value):
        return json.dumps(
            value, ensure_ascii=False, sort_keys=True,
            separators=(",", ":"), allow_nan=False,
        ).encode("utf-8")

    def sha(raw):
        return hashlib.sha256(raw).hexdigest()

    reports = Path("reports/phase2/quality_improvement/NB22E.lab.v3")
    cwd = Path.cwd()
    root = next(
        (p.resolve() for p in (cwd, *cwd.parents) if (p / reports).is_dir()),
        None,
    )
    require(root is not None, "C191_PROJECT_ROOT_NOT_FOUND")

    def read_bound(relative, digest):
        path = (root / relative).resolve()
        require(path.is_relative_to(root), "C191_PATH_OUTSIDE_PROJECT")
        raw = path.read_bytes()
        require(sha(raw) == digest, f"C191_HASH_MISMATCH:{relative}")
        return json.loads(raw)

    def report(stem, digest):
        return read_bound(reports / f"{stem}.{digest}.json", digest)

    def load_ref(pin):
        return read_bound(pin["path"], pin["file_bytes_sha256"])

    def save(stem, value):
        raw = encode(value)
        digest = sha(raw)
        relative = reports / f"{stem}.{digest}.json"
        target = root / relative
        descriptor, temporary = tempfile.mkstemp(
            prefix=".C191-", dir=str(target.parent),
        )
        try:
            with os.fdopen(descriptor, "wb") as stream:
                stream.write(raw)
                stream.flush()
                os.fsync(stream.fileno())
            try:
                os.link(temporary, target)
            except FileExistsError:
                pass
            require(target.read_bytes() == raw, "C191_OUTPUT_RELOAD_MISMATCH")
        finally:
            Path(temporary).unlink(missing_ok=True)
        return {
            "path": relative.as_posix(),
            "file_bytes_sha256": digest,
            "reload_validation": "PASS",
        }

    c190_sha = "310a9423b73324fa59d13cc78e3e499729d3e6b7682cef3c30aea0f399ba670b"
    c186_sha = "5cb0fa8851d28b1fa459f01cfbf6355fc32f1db9d5f484a133685b93d6315821"
    c156_sha = "ec4cd70abcd81fdae2659620da30edcf108d940df1e8a1bbed78fb5d7224730d"
    c151_input_sha = "63ed39d41629173f52654b679846bc5397e8dbe9a2041048cacc939223984fdd"
    c042_sha = "e8ff5073b9a0c595dda7a0cd7d49975518dac17c03c9130b9ae3c3cd35b60e8b"

    batch = report("C190.R11_batch_product_input_boundaries", c190_sha)
    inputs = report("C186.R11_remaining_scenario_input_preflight", c186_sha)
    template_plan = report("C156.R11_target_deepseek_request_preparation", c156_sha)
    template_job = load_ref(template_plan["request_receipt"])
    reference = report("C151.R11_target_generation_input", c151_input_sha)
    rates = report("C042.R11_deepseek_batch", c042_sha)

    require(
        batch["status"] == "ALL_PRODUCT_INPUT_BOUNDARIES_VERIFIED"
        and batch["verified_candidate_count"] == 33
        and batch["input_file_hashes"]["C186"] == c186_sha
        and not batch["blocked_candidates"],
        "C191_SUCCESSFUL_C190_REQUIRED",
    )
    template = template_job["request"]
    require(
        sha(encode(template)) == template_plan["request_payload_sha256"]
        == template_job["request_payload_sha256"]
        and template["model"] == "DeepSeek-V4-Flash"
        and template["temperature"] == 0
        and template["max_tokens"] == 600
        and template["thinking"] == {"type": "disabled"}
        and [message["role"] for message in template["messages"]] == ["system", "user"],
        "C191_TEMPLATE_REQUEST_MISMATCH",
    )

    old_question = reference["question"]
    old_context = reference["context"]
    user_template = template["messages"][1]
    require(
        bool(old_question) and bool(old_context)
        and user_template["content"].count(old_context) == 1
        and sha(old_context.encode("utf-8"))
        == reference["rendered_context_sha256"]
        == template_job["rendered_context_sha256"],
        "C191_REFERENCE_CONTEXT_MISMATCH",
    )
    prefix, _, suffix = user_template["content"].partition(old_context)
    require(
        prefix.count(old_question) + suffix.count(old_question) == 1,
        "C191_QUESTION_WRAPPER_AMBIGUOUS",
    )

    input_rate = rates["input_toman_per_million"]
    output_rate = rates["output_toman_per_million"]
    require(
        all(type(rate) is int and rate > 0 for rate in (input_rate, output_rate)),
        "C191_HISTORICAL_RATES_INVALID",
    )

    candidate_index = {
        item["candidate_id"]: item for item in inputs["candidate_inputs"]
    }
    expected_specs = {
        spec["regression_spec_id"]: spec for spec in inputs["scenario_specs"]
    }
    require(len(candidate_index) == 33 and len(expected_specs) == 35,
            "C191_INPUT_POPULATION_MISMATCH")

    jobs = []
    covered = set()
    seen_candidates = set()
    total_cost_units = 0

    for context_pin in batch["prepared_context_receipts"]:
        prepared = load_ref(context_pin)
        candidate = prepared["candidate_id"]
        require(
            candidate not in seen_candidates and candidate in candidate_index,
            "C191_CANDIDATE_DUPLICATE_OR_UNKNOWN",
        )
        seen_candidates.add(candidate)
        item = candidate_index[candidate]
        question = prepared["question"]
        context = prepared["rendered_context"]
        receipt = load_ref(prepared["input_receipt"])
        require(
            question == item["question"]
            and prepared["question_sha256"] == item["question_sha256"]
            == sha(question.encode("utf-8"))
            and prepared["rendered_context_sha256"]
            == receipt["rendered_context_sha256"]
            == sha(context.encode("utf-8"))
            and receipt["candidate_id"] == candidate
            and receipt["question_sha256"] == item["question_sha256"]
            and receipt["selected_variant"] == "char_rrf",
            "C191_PREPARED_CONTEXT_BINDING_MISMATCH",
        )

        specs = prepared["scenario_specs"]
        for spec in specs:
            spec_id = spec["regression_spec_id"]
            require(
                spec_id not in covered
                and expected_specs.get(spec_id) == spec
                and spec["candidate_id"] == candidate,
                "C191_SCENARIO_BINDING_MISMATCH",
            )
            covered.add(spec_id)

        request = {
            **template,
            "messages": [
                dict(template["messages"][0]),
                {
                    **user_template,
                    "content": (
                        prefix.replace(old_question, question)
                        + context
                        + suffix.replace(old_question, question)
                    ),
                },
            ],
        }
        body = encode(request)
        input_allowance = len(body) + 1024
        cost_units = input_allowance * input_rate + request["max_tokens"] * output_rate
        total_cost_units += cost_units

        job = {
            "candidate_id": candidate,
            "question_sha256": item["question_sha256"],
            "rendered_context_sha256": prepared["rendered_context_sha256"],
            "prepared_context_receipt": context_pin,
            "rank_receipt": prepared["rank_receipt"],
            "preflight_input_receipt": prepared["input_receipt"],
            "scenario_specs": specs,
            "request": request,
            "request_payload_sha256": sha(body),
            "request_bytes": len(body),
            "estimated_input_token_allowance": input_allowance,
            "historical_estimated_cost_toman": cost_units / 1_000_000,
            "dispatch_performed": False,
        }
        job_pin = save("C191.R11_prepared_deepseek_job", job)
        jobs.append({
            "candidate_id": candidate,
            "scenario_ids": [spec["regression_spec_id"] for spec in specs],
            "request_payload_sha256": job["request_payload_sha256"],
            "historical_estimated_cost_toman": job["historical_estimated_cost_toman"],
            "job_receipt": job_pin,
        })

    require(
        seen_candidates == candidate_index.keys()
        and covered == expected_specs.keys(),
        "C191_BATCH_COVERAGE_MISMATCH",
    )

    proposed_budget = 100_000
    within_budget = total_cost_units < proposed_budget * 1_000_000
    result = {
        "cell": "C191",
        "roadmap_subtask": "R11.3",
        "checklist_item": 12,
        "status": "DEEPSEEK_BATCH_PREPARED_OFFLINE",
        "input_file_hashes": {
            "C190": c190_sha,
            "C186": c186_sha,
            "C156": c156_sha,
            "C151_prepared_input": c151_input_sha,
            "C042_rate_reference": c042_sha,
        },
        "candidate_count": len(jobs),
        "scenario_count": len(covered),
        "unique_request_payload_count": len({
            job["request_payload_sha256"] for job in jobs
        }),
        "profile": {
            key: value for key, value in template.items() if key != "messages"
        },
        "jobs": jobs,
        "rate_basis": "C042_HISTORICAL_REFERENCE_RATES",
        "input_toman_per_million": input_rate,
        "output_toman_per_million": output_rate,
        "current_rate_verified": False,
        "historical_total_estimate_toman": total_cost_units / 1_000_000,
        "proposed_total_budget_toman": proposed_budget,
        "historical_estimate_within_proposed_budget": within_budget,
        "dispatch_requirement": "FRESH_INPUT_RECEIPT_AND_RESPONSE_IN_SAME_PRODUCT_INVOCATION",
        "next_dependency": (
            "RUN_RECEIPT_BOUND_PRODUCT_BATCH_WITH_ONE_ATTEMPT_PER_JOB"
            if within_budget else "RESOLVE_BATCH_BUDGET_BEFORE_DISPATCH"
        ),
        "new_api_calls": 0,
        "new_response_scenario_results": 0,
        "done_item_count": 9,
        "open_item_count": 3,
        "open_item_ids": [9, 10, 12],
        "selected_version_gate": "BLOCKED",
    }
    result["output"] = save("C191.R11_deepseek_batch_plan", result)
    print(json.dumps(
        {key: value for key, value in result.items() if key != "jobs"},
        ensure_ascii=False, indent=2,
    ))
    return result


V3_C191 = run_v3_c191()
````

### NB22E-C192 — اجرای بستهٔ DeepSeek از مسیر محصول

برای هر کاندیدا، رتبه‌های نسخه‌دار وارد ask_with_char_rrf می‌شوند.
رسید ورودی تازه پیش از تولید ذخیره و با زمینهٔ آماده‌شده تطبیق داده
می‌شود. درخواست یکتا حداکثر یک‌بار ارسال می‌شود.

درخواست‌های کاملاً یکسان می‌توانند از پاسخ ذخیره‌شده استفاده کنند.
این حالت و شناسهٔ فراخوان ارسال اصلی در رسید هر کاندیدا مشخص می‌شود.
پاسخ خام و رسید خروجی ذخیره می‌شوند تا بررسی معنایی بعدی روی همان
ورودی و پاسخ انجام شود.

برآورد هزینه از نرخ تاریخی C042 گرفته شده است. این سل ارسال API دارد.

````python
# NB22E-C192 — Execute the prepared DeepSeek batch through the product method.
# Persist fresh input receipts and send each exact request at most once.

def run_v3_c192():
    def require(ok, code):
        if not ok:
            raise ValueError(code)

    def encode(value):
        return json.dumps(
            value, ensure_ascii=False, sort_keys=True,
            separators=(",", ":"), allow_nan=False,
        ).encode("utf-8")

    def sha(raw):
        return hashlib.sha256(raw).hexdigest()

    def text_sha(text):
        return sha(text.encode("utf-8"))

    reports = Path("reports/phase2/quality_improvement/NB22E.lab.v3")
    cwd = Path.cwd()
    root = next(
        (p.resolve() for p in (cwd, *cwd.parents) if (p / reports).is_dir()),
        None,
    )
    require(root is not None, "C192_PROJECT_ROOT_NOT_FOUND")

    def project_path(relative):
        path = (root / relative).resolve()
        require(path.is_relative_to(root), "C192_PATH_OUTSIDE_PROJECT")
        return path

    def read_bound(relative, digest):
        raw = project_path(relative).read_bytes()
        require(sha(raw) == digest, f"C192_HASH_MISMATCH:{relative}")
        return json.loads(raw)

    def load_ref(pin):
        return read_bound(pin["path"], pin["file_bytes_sha256"])

    def write_once(path, raw):
        path.parent.mkdir(parents=True, exist_ok=True)
        descriptor, temporary = tempfile.mkstemp(
            prefix=".C192-", dir=str(path.parent),
        )
        try:
            with os.fdopen(descriptor, "wb") as stream:
                stream.write(raw)
                stream.flush()
                os.fsync(stream.fileno())
            try:
                os.link(temporary, path)
            except FileExistsError:
                pass
            require(path.read_bytes() == raw, "C192_EXISTING_FILE_DIFFERS")
        finally:
            Path(temporary).unlink(missing_ok=True)

    def save(stem, value):
        raw = encode(value)
        digest = sha(raw)
        relative = reports / f"{stem}.{digest}.json"
        write_once(project_path(relative), raw)
        return {
            "path": relative.as_posix(),
            "file_bytes_sha256": digest,
            "reload_validation": "PASS",
        }

    plan_sha = "a1a590e38c04ac9f0a7431bd5aac9c972f77c97f86b0e3d318c81d17d9e87876"
    plan = read_bound(
        reports / f"C191.R11_deepseek_batch_plan.{plan_sha}.json", plan_sha,
    )
    require(
        plan["status"] == "DEEPSEEK_BATCH_PREPARED_OFFLINE"
        and plan["candidate_count"] == 33
        and plan["scenario_count"] == 35
        and plan["unique_request_payload_count"] == 32
        and plan["historical_estimate_within_proposed_budget"] is True
        and plan["historical_total_estimate_toman"] < 100_000
        and plan["proposed_total_budget_toman"] == 100_000,
        "C192_PLAN_OR_BUDGET_MISMATCH",
    )

    # Validate every job before opening the product instance or sending.
    jobs = []
    candidate_ids = set()
    payloads = {}
    covered_specs = set()
    for summary in plan["jobs"]:
        job = load_ref(summary["job_receipt"])
        prepared = load_ref(job["prepared_context_receipt"])
        rank = load_ref(job["rank_receipt"])
        preflight = load_ref(prepared["input_receipt"])
        candidate = job["candidate_id"]
        body = encode(job["request"])
        request_sha = sha(body)
        require(
            candidate not in candidate_ids
            and candidate == summary["candidate_id"] == prepared["candidate_id"]
            == rank["candidate_id"] == preflight["candidate_id"]
            and job["question_sha256"] == prepared["question_sha256"]
            == text_sha(prepared["question"])
            and request_sha == summary["request_payload_sha256"]
            == job["request_payload_sha256"]
            and len(body) == job["request_bytes"]
            and {k: v for k, v in job["request"].items() if k != "messages"}
            == plan["profile"]
            and job["request"]["model"] == "DeepSeek-V4-Flash"
            and job["dispatch_performed"] is False
            and job["scenario_specs"] == prepared["scenario_specs"]
            and job["rendered_context_sha256"]
            == prepared["rendered_context_sha256"]
            == preflight["rendered_context_sha256"]
            == text_sha(prepared["rendered_context"])
            and preflight["ordered_chunk_ids"] == rank["char_rrf_top10"]
            and preflight["ordered_chunk_text_sha256"]
            == rank["char_rrf_top10_text_sha256"],
            "C192_JOB_BINDING_MISMATCH",
        )
        scenario_ids = [spec["regression_spec_id"] for spec in job["scenario_specs"]]
        require(
            not covered_specs.intersection(scenario_ids)
            and scenario_ids == summary["scenario_ids"],
            "C192_SCENARIO_DUPLICATE_OR_MISMATCH",
        )
        covered_specs.update(scenario_ids)
        candidate_ids.add(candidate)
        if request_sha in payloads:
            require(payloads[request_sha] == body, "C192_PAYLOAD_COLLISION")
        payloads[request_sha] = body
        jobs.append((summary, job, prepared, rank, preflight))

    require(
        len(jobs) == 33 and len(payloads) == 32 and len(covered_specs) == 35,
        "C192_BATCH_COVERAGE_MISMATCH",
    )

    c190_sha = plan["input_file_hashes"]["C190"]
    boundary_report = read_bound(
        reports / f"C190.R11_batch_product_input_boundaries.{c190_sha}.json",
        c190_sha,
    )
    pipeline_relative = "src/pipeline/rag_pipeline.py"
    pipeline_sha = boundary_report["input_file_hashes"]["pipeline_source"]
    pipeline_raw = project_path(pipeline_relative).read_bytes()
    require(sha(pipeline_raw) == pipeline_sha, "C192_PRODUCT_SOURCE_MISMATCH")
    if str(root) not in sys.path:
        sys.path.insert(0, str(root))
    module = importlib.import_module("src.pipeline.rag_pipeline")
    require(
        Path(module.__file__).resolve() == project_path(pipeline_relative),
        "C192_PRODUCT_MODULE_PATH_MISMATCH",
    )
    pipeline_class = module.RAGPipeline

    code_type = type(run_v3_c192.__code__)
    compiled = compile(pipeline_raw, pipeline_relative, "exec", dont_inherit=True)
    class_code = next(
        value for value in compiled.co_consts
        if isinstance(value, code_type) and value.co_name == "RAGPipeline"
    )
    method_codes = {
        value.co_name: value for value in class_code.co_consts
        if isinstance(value, code_type)
    }

    def signature(code):
        return (
            code.co_code,
            tuple(
                signature(value) if isinstance(value, code_type) else value
                for value in code.co_consts
            ),
            code.co_names, code.co_varnames, code.co_freevars, code.co_cellvars,
            code.co_flags, code.co_argcount, code.co_posonlyargcount,
            code.co_kwonlyargcount, code.co_exceptiontable,
        )

    for name in ("__init__", "ask_with_char_rrf", "build_context", "close"):
        require(
            signature(getattr(pipeline_class, name).__code__)
            == signature(method_codes[name]),
            f"C192_LOADED_METHOD_MISMATCH:{name}",
        )

    c186_sha = plan["input_file_hashes"]["C186"]
    input_batch = read_bound(
        reports / f"C186.R11_remaining_scenario_input_preflight.{c186_sha}.json",
        c186_sha,
    )
    corpus_relative = (
        "data/phase2/chunks/nb11_benchmark_v1/"
        "baseline_phase1_char_safe_v1.jsonl"
    )
    corpus_raw = project_path(corpus_relative).read_bytes()
    require(
        sha(corpus_raw) == input_batch["input_file_hashes"]["chunks"],
        "C192_CORPUS_HASH_MISMATCH",
    )
    corpus = {}
    for line in corpus_raw.splitlines():
        if line.strip():
            row = json.loads(line)
            require(row["chunk_id"] not in corpus, "C192_DUPLICATE_CHUNK")
            corpus[row["chunk_id"]] = row

    url_module = globals().get("urllib")
    parser = globals().get("dotenv_values")
    require(
        callable(parser) and url_module is not None
        and all(hasattr(url_module, name) for name in ("parse", "request", "error")),
        "C192_C001_TRANSPORT_BINDINGS_REQUIRED",
    )
    config = parser(root / ".env", interpolate=False)
    base = str(os.environ.get("ARVAN_GATEWAY_URL") or config.get("ARVAN_GATEWAY_URL") or "").strip()
    key = str(os.environ.get("ARVAN_GATEWAY_API_KEY") or config.get("ARVAN_GATEWAY_API_KEY") or "").strip()
    url = url_module.parse.urlsplit(base)
    require(
        url.scheme == "https" and bool(url.hostname) and bool(key)
        and url.username is None and url.password is None
        and not url.query and not url.fragment,
        "C192_GATEWAY_CONFIGURATION_INVALID",
    )
    route = url.path.rstrip("/") or "/v1"
    if not route.endswith("/chat/completions"):
        route += "/chat/completions"
    endpoint = url._replace(path=route).geturl()

    class NoRedirect(url_module.request.HTTPRedirectHandler):
        def redirect_request(self, req, fp, code, msg, headers, newurl):
            return None

    opener = url_module.request.build_opener(NoRedirect())
    folder = root / reports / "C192.R11_batch_receipts" / plan_sha
    folder.mkdir(parents=True, exist_ok=True)
    c148_sha = "8352cfa49bedc99c50335f57a66051a3613bdb1cf3618200dff771a8c1d654f4"
    dense = read_bound(
        reports / f"C148.R11_target_test_hybrid100_probe.{c148_sha}.json", c148_sha,
    )

    results = []
    blockers = []
    new_calls = 0
    reused_checkpoints = 0

    def verify_completed(record, candidate, job_pin, request_sha):
        require(
            record["candidate_id"] == candidate
            and record["job_receipt"] == job_pin
            and record["request_payload_sha256"] == request_sha,
            "C192_COMPLETED_JOB_BINDING_MISMATCH",
        )
        input_record = load_ref(record["input_receipt"])
        output_record = load_ref(record["output_receipt"])
        require(
            input_record["candidate_id"] == candidate
            and input_record["runtime_identity"]["invocation_id"]
            == record["invocation_id"]
            and output_record["input_receipt_file_bytes_sha256"]
            == record["input_receipt"]["file_bytes_sha256"]
            and output_record["answer_text_utf8_sha256"]
            == text_sha(record["answer"]),
            "C192_COMPLETED_OUTPUT_BINDING_MISMATCH",
        )
        response_record = load_ref(record["response_receipt"])
        require(
            response_record["request_payload_sha256"] == request_sha
            and text_sha(response_record["response_body"])
            == response_record["response_body_sha256"]
            and response_record["http_status"] == 200,
            "C192_COMPLETED_RESPONSE_BINDING_MISMATCH",
        )
        envelope = json.loads(response_record["response_body"])
        require(
            envelope["choices"][0]["message"]["content"] == record["answer"],
            "C192_COMPLETED_ANSWER_MISMATCH",
        )

    pipeline = pipeline_class(
        vectorstore_path=project_path(dense["dense_storage_path"]),
        collection_name=dense["dense_collection"],
        use_cache=False,
    )
    try:
        for summary, job, prepared, rank, preflight in jobs:
            candidate = job["candidate_id"]
            request_sha = job["request_payload_sha256"]
            job_pin = summary["job_receipt"]
            checkpoint = folder / f"{job_pin['file_bytes_sha256']}.completed.json"
            input_ref = None
            output_ref = None
            live_state = {}
            try:
                if checkpoint.exists():
                    record = json.loads(checkpoint.read_bytes())
                    verify_completed(record, candidate, job_pin, request_sha)
                    results.append(record)
                    reused_checkpoints += 1
                    continue

                invocation = sha(
                    f"C192:{candidate}:{time.time_ns()}:{os.getpid()}".encode("utf-8")
                )
                question = prepared["question"]
                chunks = [
                    {
                        "chunk_id": cid,
                        "text": corpus[cid]["content"],
                        "title": corpus[cid].get("title"),
                        "url": corpus[cid].get("url"),
                    }
                    for cid in rank["char_rrf_top10"]
                ]
                require(
                    [text_sha(chunk["text"]) for chunk in chunks]
                    == rank["char_rrf_top10_text_sha256"],
                    "C192_CHUNK_TEXT_MISMATCH",
                )
                runtime_identity = {
                    "scope": "VERSION_BOUND_STORED_RANKS_PRODUCT_GENERATION",
                    "invocation_id": invocation,
                    "pipeline_source_sha256": pipeline_sha,
                    "rank_receipt_sha256": job["rank_receipt"]["file_bytes_sha256"],
                    "job_receipt_sha256": job_pin["file_bytes_sha256"],
                    "batch_plan_sha256": plan_sha,
                }

                def rank_sources(*, question, candidate_id, top_k):
                    require(
                        candidate_id == candidate and top_k == 10
                        and text_sha(question) == job["question_sha256"],
                        "C192_RUNTIME_QUERY_MISMATCH",
                    )
                    return {
                        "candidate_id": candidate,
                        "question_sha256": job["question_sha256"],
                        "selected_variant": "char_rrf",
                        "runtime_identity": runtime_identity,
                        "ordered_chunks": chunks,
                    }

                def write_receipt(role, payload):
                    nonlocal input_ref, output_ref
                    if role == "input":
                        require(input_ref is None, "C192_DUPLICATE_INPUT")
                        input_ref = save("C192.R11_product_input", payload)
                        return input_ref["file_bytes_sha256"]
                    require(
                        role == "output" and input_ref is not None and output_ref is None
                        and payload["input_receipt_file_bytes_sha256"]
                        == input_ref["file_bytes_sha256"]
                        and payload["answer_text_utf8_sha256"] == live_state["answer_sha"],
                        "C192_OUTPUT_BINDING_MISMATCH",
                    )
                    output_ref = save("C192.R11_product_output", payload)
                    return output_ref["file_bytes_sha256"]

                class Generator:
                    def generate_answer(self, *, question, context):
                        nonlocal new_calls
                        require(input_ref is not None, "C192_INPUT_RECEIPT_REQUIRED")
                        recorded = load_ref(input_ref)
                        require(
                            question == prepared["question"]
                            and context == prepared["rendered_context"]
                            and recorded["candidate_id"] == candidate
                            and recorded["question_sha256"] == job["question_sha256"]
                            and recorded["runtime_identity"] == runtime_identity
                            and recorded["ordered_chunk_ids"] == rank["char_rrf_top10"]
                            and recorded["ordered_chunk_text_sha256"]
                            == rank["char_rrf_top10_text_sha256"]
                            and recorded["rendered_context_sha256"]
                            == text_sha(context) == job["rendered_context_sha256"],
                            "C192_PRE_GENERATION_BINDING_MISMATCH",
                        )

                        started_path = folder / f"{request_sha}.started.json"
                        response_path = folder / f"{request_sha}.response.json"
                        mode = "EXACT_REQUEST_RESPONSE_REUSE"

                        if not response_path.exists():
                            require(
                                not started_path.exists(),
                                "C192_UNKNOWN_DISPATCH_OUTCOME_NO_RESEND",
                            )
                            started = {
                                "candidate_id": candidate,
                                "invocation_id": invocation,
                                "request_payload_sha256": request_sha,
                                "job_receipt": job_pin,
                                "input_receipt": input_ref,
                                "endpoint_sha256": text_sha(endpoint),
                            }
                            # Exclusive creation prevents concurrent duplicate sends.
                            with started_path.open("xb") as stream:
                                stream.write(encode(started))
                                stream.flush()
                                os.fsync(stream.fileno())

                            outgoing = url_module.request.Request(
                                endpoint,
                                data=payloads[request_sha],
                                method="POST",
                                headers={
                                    "Authorization": f"apikey {key}",
                                    "Content-Type": "application/json",
                                },
                            )
                            new_calls += 1
                            try:
                                with opener.open(outgoing, timeout=180) as response:
                                    status = response.status
                                    response_body = response.read().decode("utf-8")
                            except url_module.error.HTTPError as error:
                                with error:
                                    status = error.code
                                    response_body = error.read().decode("utf-8")
                            except (OSError, ValueError):
                                raise ValueError("C192_UNKNOWN_DISPATCH_OUTCOME_NO_RESEND") from None

                            response_record = {
                                "request_payload_sha256": request_sha,
                                "started_receipt_sha256": sha(started_path.read_bytes()),
                                "http_status": status,
                                "response_body": response_body,
                                "response_body_sha256": text_sha(response_body),
                            }
                            write_once(response_path, encode(response_record))
                            mode = "LIVE_DISPATCH_IN_THIS_PRODUCT_INVOCATION"

                        started_raw = started_path.read_bytes()
                        started = json.loads(started_raw)
                        response_raw = response_path.read_bytes()
                        response_record = json.loads(response_raw)
                        require(
                            started["request_payload_sha256"]
                            == response_record["request_payload_sha256"] == request_sha
                            and response_record["started_receipt_sha256"] == sha(started_raw)
                            and response_record["response_body_sha256"]
                            == text_sha(response_record["response_body"])
                            and response_record["http_status"] == 200,
                            "C192_SAVED_RESPONSE_INVALID_NO_RESEND",
                        )
                        load_ref(started["job_receipt"])
                        load_ref(started["input_receipt"])
                        envelope = json.loads(response_record["response_body"])
                        choices = envelope["choices"]
                        require(len(choices) == 1, "C192_EXPECTED_ONE_CHOICE")
                        answer = choices[0]["message"]["content"]
                        require(
                            envelope["model"].strip().rsplit("/", 1)[-1].lower()
                            in {"deepseek-v4-flash", "deepseek-v4-flash-0731"}
                            and choices[0].get("finish_reason") == "stop"
                            and isinstance(answer, str) and bool(answer.strip()),
                            "C192_RESPONSE_MODEL_OR_FINISH_INVALID",
                        )
                        live_state.update(
                            answer_sha=text_sha(answer),
                            response_mode=mode,
                            dispatch_invocation_id=started["invocation_id"],
                            response_receipt={
                                "path": response_path.relative_to(root).as_posix(),
                                "file_bytes_sha256": sha(response_raw),
                            },
                            usage_as_reported=envelope.get("usage"),
                        )
                        return answer

                pipeline.generator = Generator()
                returned = pipeline.ask_with_char_rrf(
                    question,
                    candidate_id=candidate,
                    rank_sources=rank_sources,
                    write_receipt=write_receipt,
                    top_k=10,
                )
                require(
                    output_ref is not None and returned["from_cache"] is False
                    and returned["selected_variant"] == "char_rrf"
                    and returned["input_receipt_file_bytes_sha256"]
                    == input_ref["file_bytes_sha256"]
                    and returned["output_receipt_file_bytes_sha256"]
                    == output_ref["file_bytes_sha256"],
                    "C192_PRODUCT_RETURN_BINDING_MISMATCH",
                )
                record = {
                    "candidate_id": candidate,
                    "question_sha256": job["question_sha256"],
                    "job_receipt": job_pin,
                    "scenario_ids": summary["scenario_ids"],
                    "request_payload_sha256": request_sha,
                    "invocation_id": invocation,
                    "input_receipt": input_ref,
                    "output_receipt": output_ref,
                    "answer": returned["answer"],
                    **live_state,
                    "semantic_verdict": None,
                }
                verify_completed(record, candidate, job_pin, request_sha)
                write_once(checkpoint, encode(record))
                results.append(record)
                print(
                    f"C192 {len(results)}/33 complete | "
                    f"new_api_calls={new_calls} | {live_state['response_mode']}",
                    flush=True,
                )
            except Exception as error:
                blockers.append({
                    "candidate_id": candidate,
                    "type": type(error).__name__,
                    "detail": str(error),
                    "input_receipt": input_ref,
                })
                print(f"C192 blocked: {candidate} | {type(error).__name__}", flush=True)
    finally:
        pipeline.close()

    result = {
        "cell": "C192",
        "roadmap_subtask": "R11.3",
        "checklist_item": 12,
        "status": (
            "BATCH_RESPONSES_BOUND_PENDING_SEMANTIC_REVIEW"
            if len(results) == 33 and not blockers else "PRODUCT_BATCH_PARTIAL"
        ),
        "input_file_hashes": {"C191": plan_sha},
        "completed_candidate_count": len(results),
        "covered_scenario_count": len({
            spec_id for row in results for spec_id in row["scenario_ids"]
        }),
        "reused_completed_checkpoints": reused_checkpoints,
        "new_api_calls": new_calls,
        "same_invocation_live_candidate_count": sum(
            row["response_mode"] == "LIVE_DISPATCH_IN_THIS_PRODUCT_INVOCATION"
            for row in results
        ),
        "exact_request_reuse_candidate_count": sum(
            row["response_mode"] == "EXACT_REQUEST_RESPONSE_REUSE"
            for row in results
        ),
        "candidate_results": results,
        "blocked_candidates": blockers,
        "rank_input_mode": "VERSION_BOUND_STORED_C189_RANKS",
        "new_semantic_verdicts": 0,
        "rate_basis": plan["rate_basis"],
        "current_rate_verified": False,
        "historical_total_estimate_toman": plan["historical_total_estimate_toman"],
        "done_item_count": 9,
        "open_item_count": 3,
        "open_item_ids": [9, 10, 12],
        "selected_version_gate": "BLOCKED",
        "next_dependency": "ASSESS_BOUND_RESPONSES_AND_CITATION_WITNESSES",
    }
    result["output"] = save("C192.R11_product_deepseek_batch", result)
    print(json.dumps(
        {key: value for key, value in result.items() if key != "candidate_results"},
        ensure_ascii=False, indent=2,
    ))
    return result


V3_C192 = run_v3_c192()
````

````python
# NB22E-C193 — Inspect blocked responses and assemble the batch review evidence.
# Read persisted requests and responses without retrying any network operation.

def run_v3_c193():
    def require(ok, code):
        if not ok:
            raise ValueError(code)

    def encode(value):
        return json.dumps(
            value, ensure_ascii=False, sort_keys=True,
            separators=(",", ":"), allow_nan=False,
        ).encode("utf-8")

    def sha(raw):
        return hashlib.sha256(raw).hexdigest()

    def text_sha(text):
        return sha(text.encode("utf-8"))

    reports = Path("reports/phase2/quality_improvement/NB22E.lab.v3")
    cwd = Path.cwd()
    root = next(
        (p.resolve() for p in (cwd, *cwd.parents) if (p / reports).is_dir()),
        None,
    )
    require(root is not None, "C193_PROJECT_ROOT_NOT_FOUND")

    def project_path(relative):
        path = (root / relative).resolve()
        require(path.is_relative_to(root), "C193_PATH_OUTSIDE_PROJECT")
        return path

    def load(relative, digest):
        raw = project_path(relative).read_bytes()
        require(sha(raw) == digest, f"C193_HASH_MISMATCH:{relative}")
        return json.loads(raw)

    def load_ref(pin):
        return load(pin["path"], pin["file_bytes_sha256"])

    def save(stem, value):
        raw = encode(value)
        digest = sha(raw)
        relative = reports / f"{stem}.{digest}.json"
        path = project_path(relative)
        descriptor, temporary = tempfile.mkstemp(
            prefix=".C193-", dir=str(path.parent),
        )
        try:
            with os.fdopen(descriptor, "wb") as stream:
                stream.write(raw)
                stream.flush()
                os.fsync(stream.fileno())
            try:
                os.link(temporary, path)
            except FileExistsError:
                pass
            require(path.read_bytes() == raw, "C193_OUTPUT_RELOAD_MISMATCH")
        finally:
            Path(temporary).unlink(missing_ok=True)
        return {
            "path": relative.as_posix(),
            "file_bytes_sha256": digest,
            "reload_validation": "PASS",
        }

    batch_sha = "a99218d2798e7d20147caaaf75f23dc8b3ac98c1a52010c33d8c0c3d7d74a0fa"
    plan_sha = "a1a590e38c04ac9f0a7431bd5aac9c972f77c97f86b0e3d318c81d17d9e87876"
    batch = load(
        reports / f"C192.R11_product_deepseek_batch.{batch_sha}.json",
        batch_sha,
    )
    plan = load(
        reports / f"C191.R11_deepseek_batch_plan.{plan_sha}.json",
        plan_sha,
    )
    require(
        batch["input_file_hashes"]["C191"] == plan_sha
        and batch["completed_candidate_count"] == 31
        and len(batch["blocked_candidates"]) == 2,
        "C193_BATCH_BINDING_MISMATCH",
    )

    completed = {row["candidate_id"]: row for row in batch["candidate_results"]}
    blocked = {row["candidate_id"]: row for row in batch["blocked_candidates"]}
    require(
        len(completed) == 31 and len(blocked) == 2
        and not completed.keys() & blocked.keys(),
        "C193_CANDIDATE_PARTITION_MISMATCH",
    )
    folder = reports / "C192.R11_batch_receipts" / plan_sha
    reviews = []
    issues = []

    for summary in plan["jobs"]:
        candidate = summary["candidate_id"]
        job = load_ref(summary["job_receipt"])
        prepared = load_ref(job["prepared_context_receipt"])
        request_sha = sha(encode(job["request"]))
        require(
            request_sha == summary["request_payload_sha256"]
            == job["request_payload_sha256"]
            and job["candidate_id"] == prepared["candidate_id"] == candidate,
            "C193_JOB_IDENTITY_MISMATCH",
        )

        response_relative = folder / f"{request_sha}.response.json"
        response_raw = project_path(response_relative).read_bytes()
        response = json.loads(response_raw)
        started = load(
            folder / f"{request_sha}.started.json",
            response["started_receipt_sha256"],
        )
        original_job = load_ref(started["job_receipt"])
        original_input = load_ref(started["input_receipt"])
        require(
            response["request_payload_sha256"]
            == started["request_payload_sha256"] == request_sha
            and sha(encode(original_job["request"])) == request_sha
            and original_input["candidate_id"] == started["candidate_id"]
            and original_input["runtime_identity"]["invocation_id"]
            == started["invocation_id"]
            and response["response_body_sha256"]
            == text_sha(response["response_body"])
            and response["http_status"] == 200,
            "C193_RESPONSE_DISPATCH_CHAIN_MISMATCH",
        )

        current = completed.get(candidate)
        input_pin = current["input_receipt"] if current else blocked[candidate]["input_receipt"]
        input_record = load_ref(input_pin)
        require(
            input_record["candidate_id"] == candidate
            and input_record["question_sha256"] == job["question_sha256"]
            and input_record["rendered_context_sha256"]
            == prepared["rendered_context_sha256"]
            == text_sha(prepared["rendered_context"]),
            "C193_CURRENT_INPUT_BINDING_MISMATCH",
        )

        envelope = json.loads(response["response_body"])
        choices = envelope.get("choices", [])
        choice = choices[0] if len(choices) == 1 else {}
        message = choice.get("message", {})
        answer = message.get("content")
        finish = choice.get("finish_reason")
        model = str(envelope.get("model", "")).strip().rsplit("/", 1)[-1].lower()
        model_ok = model in {"deepseek-v4-flash", "deepseek-v4-flash-0731"}
        text_ok = isinstance(answer, str) and bool(answer.strip())

        if current:
            output_record = load_ref(current["output_receipt"])
            require(
                current["job_receipt"] == summary["job_receipt"]
                and current["request_payload_sha256"] == request_sha
                and current["response_receipt"]["file_bytes_sha256"] == sha(response_raw)
                and current["answer"] == answer
                and output_record["answer_text_utf8_sha256"] == text_sha(answer)
                and output_record["input_receipt_file_bytes_sha256"]
                == input_pin["file_bytes_sha256"]
                and model_ok and text_ok and finish == "stop",
                "C193_COMPLETED_RESULT_MISMATCH",
            )

        reasons = []
        if not model_ok:
            reasons.append("RESPONSE_MODEL_MISMATCH")
        if len(choices) != 1:
            reasons.append("CHOICE_COUNT_MISMATCH")
        if finish != "stop":
            reasons.append(
                "OUTPUT_TOKEN_LIMIT_REACHED"
                if finish == "length" else "FINISH_REASON_NOT_STOP"
            )
        if not text_ok:
            reasons.append("EMPTY_OR_NON_TEXT_CONTENT")

        if not current:
            issues.append({
                "candidate_id": candidate,
                "scenario_ids": summary["scenario_ids"],
                "response_model": model,
                "finish_reason": finish,
                "content_type": type(answer).__name__,
                "content_characters": len(answer) if isinstance(answer, str) else None,
                "usage_as_reported": envelope.get("usage"),
                "diagnosed_reasons": reasons,
                "response_receipt": {
                    "path": response_relative.as_posix(),
                    "file_bytes_sha256": sha(response_raw),
                },
            })

        reviews.append({
            "candidate_id": candidate,
            "question": prepared["question"],
            "question_sha256": job["question_sha256"],
            "scenario_specs": job["scenario_specs"],
            "request_payload_sha256": request_sha,
            "input_receipt": input_pin,
            "rendered_context_sha256": prepared["rendered_context_sha256"],
            "exact_dispatched_context": prepared["rendered_context"],
            "response_model": model,
            "finish_reason": finish,
            "answer": answer,
            "answer_text_sha256": text_sha(answer) if isinstance(answer, str) else None,
            "usage_as_reported": envelope.get("usage"),
            "product_output_receipt": current["output_receipt"] if current else None,
            "response_receipt": {
                "path": response_relative.as_posix(),
                "file_bytes_sha256": sha(response_raw),
            },
            "execution_status": "COMPLETE" if current else "REQUIRES_REVIEW",
            "diagnosed_reasons": reasons,
            "semantic_verdict": None,
        })

    require(
        len(reviews) == 33
        and len({row["candidate_id"] for row in reviews}) == 33
        and len({
            spec["regression_spec_id"]
            for row in reviews for spec in row["scenario_specs"]
        }) == 35,
        "C193_REVIEW_COVERAGE_MISMATCH",
    )
    packet_pin = save("C193.R11_batch_semantic_review_packet", {
        "cell": "C193",
        "scope": "33_CANDIDATES_35_REMAINING_SCENARIOS",
        "input_file_hashes": {"C192": batch_sha, "C191": plan_sha},
        "candidate_reviews": reviews,
        "new_semantic_verdicts": 0,
    })

    result = {
        "cell": "C193",
        "roadmap_subtask": "R11.3",
        "checklist_item": 12,
        "status": "BATCH_REVIEW_PACKET_AND_BLOCKED_RESPONSE_DIAGNOSIS_READY",
        "input_file_hashes": {"C192": batch_sha, "C191": plan_sha},
        "candidate_count": 33,
        "scenario_count": 35,
        "completed_product_results": 31,
        "blocked_response_diagnoses": issues,
        "review_packet": packet_pin,
        "new_api_calls": 0,
        "new_semantic_verdicts": 0,
        "done_item_count": 9,
        "open_item_count": 3,
        "open_item_ids": [9, 10, 12],
        "selected_version_gate": "BLOCKED",
    }
    result["output"] = save("C193.R11_batch_response_diagnosis", result)
    print(json.dumps(result, ensure_ascii=False, indent=2))
    return result


V3_C193 = run_v3_c193()
````

### NB22E-C194 — ثبت تجمیعی انطباق رفتاری ۳۵ سناریو

````python
# NB22E-C194 — Record behavioral conformity for all remaining scenarios.
# Bind observations to exact response and context hashes while retaining review gaps.

def run_v3_c194():
    def require(ok, code):
        if not ok:
            raise ValueError(code)

    def sha(raw):
        return hashlib.sha256(raw).hexdigest()

    def text_sha(text):
        return sha(text.encode("utf-8"))

    reports = Path("reports/phase2/quality_improvement/NB22E.lab.v3")
    cwd = Path.cwd()
    root = next(
        (path.resolve() for path in (cwd, *cwd.parents) if (path / reports).is_dir()),
        None,
    )
    require(root is not None, "C194_PROJECT_ROOT_NOT_FOUND")

    def load_report(stem, digest):
        path = root / reports / f"{stem}.{digest}.json"
        raw = path.read_bytes()
        require(sha(raw) == digest, f"C194_HASH_MISMATCH:{stem}")
        return json.loads(raw)

    packet_sha = "0b20b53e6812c69c47809a027f2c33c59e285bc6baf509db91d4e2707f460a4b"
    c185_sha = "33902850e52c924fd8f2d8ae63e2256586b91141b2e6ae46574671aecc59a6a0"
    packet = load_report("C193.R11_batch_semantic_review_packet", packet_sha)
    prior = load_report("C185.R11_first_live_scenario_result", c185_sha)

    reviewed_clarification_cases = {
        "REG_0a54edaa013a5a61bd1b64c5": "پاسخ فقط اعلام ناکافی‌بودن منابع است.",
        "REG_14fca129f339001476335a7f": "نوع صندوق بدون درخواست توضیح، صندوق سرمایه‌گذاری فرض شده است.",
        "REG_b3c377fa1b73c9f527d6c465": "مرجع «این موضوع» روشن نشده و پاسخ مقایسه را تأیید کرده است.",
        "REG_986fe35346397b1487b7ac7f": "پاسخ فقط اعلام ناکافی‌بودن منابع است.",
        "REG_9651cf3bedcdf561a275c124": "مرجع «این‌ها» روشن نشده و پاسخ فهرستی از اقدامات ارائه کرده است.",
    }
    bare_abstention = "در منابع موجود پاسخ کافی پیدا نشد."
    counts = {
        "BARE_ABSTENTION_EXPECTATION_MISMATCH": 0,
        "CLARIFICATION_NOT_REQUESTED": 0,
        "NO_COMPLETED_RESPONSE": 0,
        "SUBSTANTIVE_ANSWER_REVIEW_PENDING": 0,
        "SELECTED_CLAIM_LINK_WITNESS_REVIEW_PENDING": 0,
    }
    rows = []
    seen_specs = set()
    seen_candidates = set()

    for candidate in packet["candidate_reviews"]:
        candidate_id = candidate["candidate_id"]
        require(candidate_id not in seen_candidates, "C194_DUPLICATE_CANDIDATE")
        seen_candidates.add(candidate_id)
        answer = candidate["answer"]
        context = candidate["exact_dispatched_context"]
        require(
            text_sha(candidate["question"]) == candidate["question_sha256"]
            and text_sha(context) == candidate["rendered_context_sha256"],
            "C194_QUESTION_OR_CONTEXT_HASH_MISMATCH",
        )
        if isinstance(answer, str):
            require(
                text_sha(answer) == candidate["answer_text_sha256"],
                "C194_ANSWER_HASH_MISMATCH",
            )
        else:
            require(
                answer is None and candidate["answer_text_sha256"] is None,
                "C194_UNSUPPORTED_RESPONSE_TYPE",
            )

        for spec in candidate["scenario_specs"]:
            spec_id = spec["regression_spec_id"]
            require(
                spec_id not in seen_specs and spec["candidate_id"] == candidate_id,
                "C194_SCENARIO_IDENTITY_MISMATCH",
            )
            seen_specs.add(spec_id)
            expected = spec.get("behavior_expectation", {}).get("expected_behavior")
            reason = None

            if answer is None:
                require(
                    candidate["finish_reason"] == "length"
                    and candidate["execution_status"] == "REQUIRES_REVIEW",
                    "C194_UNEXPECTED_EMPTY_RESPONSE_STATE",
                )
                observation = "NO_COMPLETED_RESPONSE"
                reason = "تولید با سقف توکن متوقف شده و متن پاسخ موجود نیست."
            elif spec["kind"] == "CITATION_WITNESS":
                observation = "SELECTED_CLAIM_LINK_WITNESS_REVIEW_PENDING"
                reason = "ادعای پاسخ جدید و پیوند آن باید با شاهد منتخب بررسی شود."
            elif spec_id in reviewed_clarification_cases:
                require(
                    expected == "CLARIFICATION_REQUIRED",
                    "C194_CLARIFICATION_EXPECTATION_CHANGED",
                )
                observation = "CLARIFICATION_NOT_REQUESTED"
                reason = reviewed_clarification_cases[spec_id]
            elif answer.strip() == bare_abstention:
                require(
                    expected in {"ANSWER_SUPPORTED", "BOUNDED_ANSWER_SUPPORTED"},
                    "C194_UNEXPECTED_ABSTENTION_EXPECTATION",
                )
                observation = "BARE_ABSTENTION_EXPECTATION_MISMATCH"
                reason = "انتظار دفتر پاسخ است، اما متن فقط اعلام ناکافی‌بودن منابع است."
            else:
                require(
                    spec["kind"] == "ANSWER_DECISION",
                    "C194_UNSUPPORTED_SCENARIO_KIND",
                )
                observation = "SUBSTANTIVE_ANSWER_REVIEW_PENDING"
                reason = "پاسخ محتوایی یا توضیح‌دار است و داوری ادعاها لازم است."

            counts[observation] += 1
            mismatch = observation in {
                "BARE_ABSTENTION_EXPECTATION_MISMATCH",
                "CLARIFICATION_NOT_REQUESTED",
            }
            rows.append({
                "regression_spec_id": spec_id,
                "candidate_id": candidate_id,
                "scenario_kind": spec["kind"],
                "expected_behavior": expected,
                "question_sha256": candidate["question_sha256"],
                "rendered_context_sha256": candidate["rendered_context_sha256"],
                "request_payload_sha256": candidate["request_payload_sha256"],
                "response_text_sha256": candidate["answer_text_sha256"],
                "input_receipt": candidate["input_receipt"],
                "response_receipt": candidate["response_receipt"],
                "product_output_receipt": candidate["product_output_receipt"],
                "observation": observation,
                "reason_fa": reason,
                "recorded_expectation_met": False if mismatch else None,
                "reference_scope_review_required": mismatch,
                "final_semantic_verdict": None,
            })

    require(
        len(seen_candidates) == 33 and len(rows) == len(seen_specs) == 35
        and counts == {
            "BARE_ABSTENTION_EXPECTATION_MISMATCH": 11,
            "CLARIFICATION_NOT_REQUESTED": 5,
            "NO_COMPLETED_RESPONSE": 2,
            "SUBSTANTIVE_ANSWER_REVIEW_PENDING": 12,
            "SELECTED_CLAIM_LINK_WITNESS_REVIEW_PENDING": 5,
        },
        "C194_REVIEWED_POPULATION_MISMATCH",
    )
    require(
        prior["scenario_result"]["verdict"] == "PASS_SCOPED_EXPECTED_BEHAVIOR"
        and prior["scenario_result"]["regression_spec_id"] not in seen_specs,
        "C194_PRESERVED_SAVINGS_RESULT_MISMATCH",
    )

    result = {
        "cell": "C194",
        "roadmap_subtask": "R11.3",
        "checklist_item": 12,
        "status": "BATCH_BEHAVIOR_OBSERVATIONS_BOUND_TO_SCENARIOS",
        "input_file_hashes": {"C193_review_packet": packet_sha, "C185": c185_sha},
        "remaining_batch_scenario_count": 35,
        "observation_counts": counts,
        "recorded_expectation_mismatch_count": 16,
        "reference_scope_review_required_count": 16,
        "preserved_savings_result": prior["scenario_result"],
        "scenario_observations": rows,
        "new_api_calls": 0,
        "new_final_semantic_verdicts": 0,
        "done_item_count": 9,
        "open_item_count": 3,
        "open_item_ids": [9, 10, 12],
        "selected_version_gate": "BLOCKED",
        "next_dependency": "ADJUDICATE_CONTENT_AND_CITATION_CASES_AND_REFERENCE_SCOPE",
    }

    raw = (
        json.dumps(result, ensure_ascii=False, sort_keys=True, indent=2) + "\n"
    ).encode("utf-8")
    digest = sha(raw)
    relative = reports / f"C194.R11_batch_behavior_observations.{digest}.json"
    path = root / relative
    try:
        with path.open("xb") as stream:
            stream.write(raw)
    except FileExistsError:
        pass
    require(path.read_bytes() == raw, "C194_OUTPUT_RELOAD_MISMATCH")
    result["output"] = {
        "path": relative.as_posix(),
        "file_bytes_sha256": digest,
        "reload_validation": "PASS",
    }
    print(json.dumps(
        {key: value for key, value in result.items()
         if key not in {"scenario_observations", "preserved_savings_result"}},
        ensure_ascii=False, indent=2,
    ))
    return result


V3_C194 = run_v3_c194()
````

````python
# NB22E-C198 — Record explicit dispositions for the five citation witness cases.
# Bind selected claim spans and limited witnesses without inventing citation links.

def run_v3_c198():
    def require(ok, code):
        if not ok:
            raise ValueError(code)

    def sha(raw):
        return hashlib.sha256(raw).hexdigest()

    reports = Path("reports/phase2/quality_improvement/NB22E.lab.v3")
    cwd = Path.cwd()
    root = next(
        (p.resolve() for p in (cwd, *cwd.parents) if (p / reports).is_dir()),
        None,
    )
    require(root is not None, "C198_PROJECT_ROOT_NOT_FOUND")

    def load(stem, digest):
        raw = (root / reports / f"{stem}.{digest}.json").read_bytes()
        require(sha(raw) == digest, f"C198_HASH_MISMATCH:{stem}")
        return json.loads(raw)

    packet_sha = "0b20b53e6812c69c47809a027f2c33c59e285bc6baf509db91d4e2707f460a4b"
    prior_sha = "18be0f15f0bcf206f4bc2101b6e4f614b2b769b83a5e37ba48e8780667b7eff8"
    packet = load("C193.R11_batch_semantic_review_packet", packet_sha)
    prior = load("C197.R11_batch_content_disposition", prior_sha)
    require(
        prior["input_file_hashes"]["C193_review_packet"] == packet_sha,
        "C198_PARENT_PACKET_MISMATCH",
    )
    candidates = {
        row["candidate_id"]: row for row in packet["candidate_reviews"]
    }

    def parse_sources(context):
        sources = {}
        for match in re.finditer(
            r"\[منبع (\d+)\]\n(.*?)(?=\n\[منبع \d+\]\n|\Z)",
            context, flags=re.S,
        ):
            chunk = re.search(r"شناسه چانک: ([^\n]+)", match.group(2))
            require(chunk is not None, "C198_CHUNK_ID_MISSING")
            cid = chunk.group(1).strip()
            require(cid not in sources, "C198_DUPLICATE_SOURCE_CHUNK")
            sources[cid] = {
                "number": int(match.group(1)),
                "match": match,
            }
        return sources

    def context_witness(context, source, quote):
        match = source["match"]
        require(match.group(2).count(quote) == 1, "C198_QUOTE_NOT_UNIQUE")
        start = context.index(quote, match.start(), match.end())
        chunk = re.search(r"شناسه چانک: ([^\n]+)", match.group(2))
        return {
            "source_number": source["number"],
            "chunk_id": chunk.group(1).strip(),
            "exact_quote": quote,
            "offset_reference": "C193_EXACT_DISPATCHED_CONTEXT",
            "span_unit": "UNICODE_CODE_POINT",
            "start": start,
            "end": start + len(quote),
        }

    def answer_span(answer, quote):
        require(answer.count(quote) == 1, "C198_SELECTED_CLAIM_NOT_UNIQUE")
        start = answer.index(quote)
        return {
            "text": quote,
            "offset_reference": "SELECTED_RESPONSE_TEXT",
            "span_unit": "UNICODE_CODE_POINT",
            "start": start,
            "end": start + len(quote),
        }

    personal_source = "chunk:baseline_phase1_char_safe:retrieval:bd123f42cf8bc714ca160a86"
    absent_source = "chunk:baseline_phase1_char_safe:retrieval:803fdde1444aa5c32c981a58"
    company_source = "chunk:baseline_phase1_char_safe:retrieval:7edaa929dc4fcc7c4d9edb76"
    assembly_source = "chunk:baseline_phase1_char_safe:retrieval:1b52697ab290fb1347de2300"

    rows = []
    reviewed_ids = []
    outcomes = {}
    for old in prior["scenario_dispositions"]:
        row = dict(old)
        if old["scenario_kind"] == "CITATION_WITNESS":
            require(
                old["scenario_disposition"]
                == "BLOCKED_SELECTED_CLAIM_LINK_NOT_ESTABLISHED",
                "C198_UNEXPECTED_CITATION_PARENT_STATUS",
            )
            candidate = candidates[row["candidate_id"]]
            context = candidate["exact_dispatched_context"]
            answer = candidate["answer"]
            require(
                sha(context.encode("utf-8")) == row["rendered_context_sha256"]
                and sha(answer.encode("utf-8")) == row["response_text_sha256"],
                "C198_SELECTED_RESPONSE_BINDING_MISMATCH",
            )
            specs = [
                spec for spec in candidate["scenario_specs"]
                if spec["regression_spec_id"] == row["regression_spec_id"]
            ]
            require(len(specs) == 1, "C198_SPEC_NOT_UNIQUE")
            historical = specs[0]["historical_link_expectation"]
            evidence_id = historical["evidence_id"]
            sources = parse_sources(context)

            audit = {
                "historical_evidence_id": evidence_id,
                "historical_link_expectation": historical,
                "historical_offsets_reused": False,
                "actual_selected_citation_link_verified": False,
                "supporting_link_pass_granted": False,
                "witnesses": [],
            }

            if evidence_id == absent_source:
                require(evidence_id not in sources, "C198_EXPECTED_ABSENT_SOURCE_PRESENT")
                status = "NOT_APPLICABLE_HISTORICAL_SOURCE_ABSENT"
                audit.update(
                    historical_link_condition="SOURCE_ABSENT_FROM_SELECTED_CONTEXT",
                    explicit_link_verdict="HISTORICAL_LINK_NOT_RETAINABLE_IN_THIS_INPUT",
                    reason_fa="قطعهٔ پیوند تاریخی در زمینهٔ منتخب وجود ندارد؛ برای آن شاهد منتخب ساخته نمی‌شود.",
                )
            elif evidence_id == personal_source:
                require(evidence_id in sources, "C198_PERSONAL_SOURCE_MISSING")
                selected_claim = answer_span(
                    answer,
                    "چون کاملاً امری شخصی است و نسخهٔ کلی برای همه یکسان نیست.",
                )
                audit.update(
                    selected_claim_span=selected_claim,
                    content_support="SUPPORTED_FOR_PERSONAL_NATURE_SUBCLAIM_ONLY",
                    explicit_link_verdict="UNDETERMINED_ACTUAL_SELECTED_LINK",
                    reason_fa="منبع از شخصی‌بودن مدیریت مالی پشتیبانی می‌کند؛ تعریف و تمام حوزه‌های ادعای مرکب از همین شاهد به‌تنهایی احراز نمی‌شوند.",
                    witnesses=[
                        context_witness(
                            context, sources[evidence_id],
                            "مدیریت مالی شخصی، شخصیه.",
                        )
                    ],
                )
                status = "EXPLICIT_UNDETERMINED_SELECTED_LINK"
            elif evidence_id == company_source:
                require(
                    evidence_id in sources and assembly_source in sources,
                    "C198_COMPANY_SOURCES_MISSING",
                )
                audit.update(
                    selected_claim_span=answer_span(
                        answer,
                        "هیئت مدیره مدیرعامل را منصوب می‌کند",
                    ),
                    content_support="PARTIAL_SUPPORT_FOR_MANAGEMENT_SUBCLAIM",
                    explicit_link_verdict="UNDETERMINED_ACTUAL_SELECTED_LINK",
                    reason_fa="منبع ۱ بخشی از توضیح مدیریت را پشتیبانی می‌کند؛ پیوند واقعی ادعای مرکب به شاهد منتخب احراز نشده است.",
                    witnesses=[
                        context_witness(
                            context, sources[evidence_id],
                            "با اختیارات معین، شرکت را مدیریت کرده و در چارچوب شرح وظایف خود تصمیم‌گیری می‌کنند.",
                        )
                    ],
                    additional_detail_issue={
                        "selected_claim_span": answer_span(
                            answer,
                            "تقسیم سود نیز با پیشنهاد مدیریت و تصویب هیئت مدیره انجام می‌شود.",
                        ),
                        "assessment": "PROFIT_APPROVAL_ROLE_DISTINCTION_OMITTED",
                        "witness": context_witness(
                            context, sources[assembly_source],
                            "میزان سود نقدی تقسیمی بین سهام‌داران تصویب می‌شود.",
                        ),
                    },
                )
                status = "EXPLICIT_UNDETERMINED_SELECTED_LINK"
            else:
                raise ValueError("C198_UNREVIEWED_HISTORICAL_LINK")

            row.update(
                scenario_disposition=status,
                citation_witness_audit=audit,
                final_semantic_verdict=None,
            )
            reviewed_ids.append(row["regression_spec_id"])
            outcomes[status] = outcomes.get(status, 0) + 1
        rows.append(row)

    require(
        len(reviewed_ids) == len(set(reviewed_ids)) == 5
        and outcomes == {
            "NOT_APPLICABLE_HISTORICAL_SOURCE_ABSENT": 2,
            "EXPLICIT_UNDETERMINED_SELECTED_LINK": 3,
        },
        "C198_CITATION_REVIEW_COUNT_MISMATCH",
    )
    counts = {}
    for row in rows:
        status = row["scenario_disposition"]
        counts[status] = counts.get(status, 0) + 1

    result = {
        "cell": "C198",
        "roadmap_subtask": "R11.3",
        "checklist_item": 12,
        "status": "FIVE_CITATION_CASES_HAVE_EXPLICIT_WITNESS_DISPOSITIONS",
        "input_file_hashes": {"C193_review_packet": packet_sha, "C197": prior_sha},
        "reviewed_citation_scenario_ids": reviewed_ids,
        "citation_disposition_counts": outcomes,
        "new_supporting_citation_link_passes": 0,
        "disposition_counts": counts,
        "scenario_dispositions": rows,
        "preserved_savings_result": prior["preserved_savings_result"],
        "total_scoped_passes_including_savings": 8,
        "formal_component_nonregression_status": "NOT_COMPARABLE_NOT_PASS",
        "new_api_calls": 0,
        "done_item_count": 9,
        "open_item_count": 3,
        "open_item_ids": [9, 10, 12],
        "selected_version_gate": "BLOCKED",
    }

    raw = (
        json.dumps(result, ensure_ascii=False, sort_keys=True, indent=2) + "\n"
    ).encode("utf-8")
    digest = sha(raw)
    relative = reports / f"C198.R11_selected_citation_witness_disposition.{digest}.json"
    path = root / relative
    try:
        with path.open("xb") as stream:
            stream.write(raw)
    except FileExistsError:
        pass
    require(path.read_bytes() == raw, "C198_OUTPUT_RELOAD_MISMATCH")
    result["output"] = {
        "path": relative.as_posix(),
        "file_bytes_sha256": digest,
        "reload_validation": "PASS",
    }
    print(json.dumps(
        {
            key: value for key, value in result.items()
            if key not in {"scenario_dispositions", "preserved_savings_result"}
        },
        ensure_ascii=False, indent=2,
    ))
    return result


V3_C198 = run_v3_c198()
````

````python
# NB22E-C199 — Probe a larger output cap for the two incomplete responses.
# Change only max_tokens and preserve the original 600-token execution evidence.

def run_v3_c199():
    def require(ok, code):
        if not ok:
            raise ValueError(code)

    def encode(value):
        return json.dumps(
            value, ensure_ascii=False, sort_keys=True,
            separators=(",", ":"), allow_nan=False,
        ).encode("utf-8")

    def sha(raw):
        return hashlib.sha256(raw).hexdigest()

    def text_sha(text):
        return sha(text.encode("utf-8"))

    reports = Path("reports/phase2/quality_improvement/NB22E.lab.v3")
    cwd = Path.cwd()
    root = next(
        (p.resolve() for p in (cwd, *cwd.parents) if (p / reports).is_dir()),
        None,
    )
    require(root is not None, "C199_PROJECT_ROOT_NOT_FOUND")

    def project_path(relative):
        path = (root / relative).resolve()
        require(path.is_relative_to(root), "C199_PATH_OUTSIDE_PROJECT")
        return path

    def load(relative, digest):
        raw = project_path(relative).read_bytes()
        require(sha(raw) == digest, f"C199_HASH_MISMATCH:{relative}")
        return json.loads(raw)

    def load_ref(pin):
        return load(pin["path"], pin["file_bytes_sha256"])

    def report(stem, digest):
        return load(reports / f"{stem}.{digest}.json", digest)

    def atomic_create(path, raw):
        path.parent.mkdir(parents=True, exist_ok=True)
        descriptor, temporary = tempfile.mkstemp(
            prefix=".C199-", dir=str(path.parent),
        )
        try:
            with os.fdopen(descriptor, "wb") as stream:
                stream.write(raw)
                stream.flush()
                os.fsync(stream.fileno())
            try:
                os.link(temporary, path)
            except FileExistsError:
                pass
            require(path.read_bytes() == raw, "C199_EXISTING_OUTPUT_DIFFERS")
        finally:
            Path(temporary).unlink(missing_ok=True)

    def save(stem, value):
        raw = encode(value)
        digest = sha(raw)
        relative = reports / f"{stem}.{digest}.json"
        atomic_create(project_path(relative), raw)
        return {
            "path": relative.as_posix(),
            "file_bytes_sha256": digest,
            "reload_validation": "PASS",
        }

    c198_sha = "e9c605ecf404f084c0953b00f01e5d8dcb7752885aa720ffa83bf19c7789e6a3"
    c193_sha = "2f0c1535a4b0f5bd264d3ceddddc8a0ae235902c1deba5978fbe7479e8b4eeff"
    plan_sha = "a1a590e38c04ac9f0a7431bd5aac9c972f77c97f86b0e3d318c81d17d9e87876"
    ledger = report("C198.R11_selected_citation_witness_disposition", c198_sha)
    diagnosis = report("C193.R11_batch_response_diagnosis", c193_sha)
    plan = report("C191.R11_deepseek_batch_plan", plan_sha)

    target_ids = {
        "golden_v2_candidate_b8756d70dae44e64533a6691",
        "golden_v2_candidate_e5ea19069f2d84ca861c2847",
    }
    diagnosed = {
        row["candidate_id"]: row
        for row in diagnosis["blocked_response_diagnoses"]
    }
    require(
        set(diagnosed) == target_ids
        and ledger["open_item_ids"] == [9, 10, 12],
        "C199_TARGET_OR_LEDGER_MISMATCH",
    )

    jobs = []
    new_estimate_units = 0
    for summary in plan["jobs"]:
        candidate = summary["candidate_id"]
        if candidate not in target_ids:
            continue
        old_job = load_ref(summary["job_receipt"])
        old_response = load_ref(diagnosed[candidate]["response_receipt"])
        old_envelope = json.loads(old_response["response_body"])
        old_choice = old_envelope["choices"][0]
        old_request = old_job["request"]
        require(
            old_response["http_status"] == 200
            and old_response["request_payload_sha256"]
            == sha(encode(old_request)) == summary["request_payload_sha256"]
            and text_sha(old_response["response_body"])
            == old_response["response_body_sha256"]
            and old_choice["finish_reason"] == "length"
            and old_choice["message"]["content"] is None
            and old_request["max_tokens"] == 600,
            "C199_ORIGINAL_INCOMPLETE_RESPONSE_MISMATCH",
        )

        prepared = load_ref(old_job["prepared_context_receipt"])
        rank = load_ref(old_job["rank_receipt"])
        request = {**old_request, "max_tokens": 2400}
        body = encode(request)
        require(
            prepared["candidate_id"] == rank["candidate_id"] == candidate
            and {k: v for k, v in request.items() if k != "max_tokens"}
            == {k: v for k, v in old_request.items() if k != "max_tokens"}
            and request["model"] == "DeepSeek-V4-Flash",
            "C199_PROFILE_CHANGE_MISMATCH",
        )
        cost_units = (
            (len(body) + 1024) * plan["input_toman_per_million"]
            + 2400 * plan["output_toman_per_million"]
        )
        new_estimate_units += cost_units
        job = {
            **old_job,
            "request": request,
            "request_payload_sha256": sha(body),
            "request_bytes": len(body),
            "original_job_receipt": summary["job_receipt"],
            "original_response_receipt": diagnosed[candidate]["response_receipt"],
            "original_request_payload_sha256": summary["request_payload_sha256"],
            "profile_change": {"max_tokens": {"before": 600, "after": 2400}},
            "historical_estimated_cost_toman": cost_units / 1_000_000,
            "dispatch_performed": False,
        }
        pin = save("C199.R11_output_cap_probe_job", job)
        jobs.append((job, pin, prepared, rank))

    require(len(jobs) == 2, "C199_EXPECTED_TWO_JOBS")
    combined_estimate = (
        plan["historical_total_estimate_toman"] + new_estimate_units / 1_000_000
    )
    require(combined_estimate < 100_000, "C199_COMBINED_ESTIMATE_REACHES_BUDGET")

    batch_plan = {
        "C191": plan_sha,
        "C198": c198_sha,
        "jobs": [pin for _, pin, _, _ in jobs],
        "historical_new_estimate_toman": new_estimate_units / 1_000_000,
        "historical_combined_estimate_toman": combined_estimate,
        "current_rate_verified": False,
        "budget_toman": 100_000,
    }
    probe_plan_pin = save("C199.R11_output_cap_probe_plan", batch_plan)
    folder = (
        root / reports / "C199.R11_output_cap_receipts"
        / probe_plan_pin["file_bytes_sha256"]
    )
    folder.mkdir(parents=True, exist_ok=True)

    c186_sha = plan["input_file_hashes"]["C186"]
    inputs = report("C186.R11_remaining_scenario_input_preflight", c186_sha)
    corpus_relative = (
        "data/phase2/chunks/nb11_benchmark_v1/baseline_phase1_char_safe_v1.jsonl"
    )
    corpus_raw = project_path(corpus_relative).read_bytes()
    require(
        sha(corpus_raw) == inputs["input_file_hashes"]["chunks"],
        "C199_CORPUS_HASH_MISMATCH",
    )
    corpus = {}
    for line in corpus_raw.splitlines():
        if line.strip():
            row = json.loads(line)
            require(row["chunk_id"] not in corpus, "C199_DUPLICATE_CHUNK")
            corpus[row["chunk_id"]] = row

    pipeline_relative = "src/pipeline/rag_pipeline.py"
    pipeline_sha = "99ee048d00a7ad1884540fa297eb41df86ad4e54f6c49f26385ebb6b2ed504f6"
    pipeline_raw = project_path(pipeline_relative).read_bytes()
    require(sha(pipeline_raw) == pipeline_sha, "C199_PIPELINE_HASH_MISMATCH")
    if str(root) not in sys.path:
        sys.path.insert(0, str(root))
    module = importlib.import_module("src.pipeline.rag_pipeline")
    require(
        Path(module.__file__).resolve() == project_path(pipeline_relative),
        "C199_PRODUCT_MODULE_PATH_MISMATCH",
    )
    pipeline_class = module.RAGPipeline

    code_type = type(run_v3_c199.__code__)
    compiled = compile(pipeline_raw, pipeline_relative, "exec", dont_inherit=True)
    class_code = next(
        value for value in compiled.co_consts
        if isinstance(value, code_type) and value.co_name == "RAGPipeline"
    )
    method_codes = {
        value.co_name: value for value in class_code.co_consts
        if isinstance(value, code_type)
    }

    def signature(code):
        return (
            code.co_code,
            tuple(
                signature(value) if isinstance(value, code_type) else value
                for value in code.co_consts
            ),
            code.co_names, code.co_varnames, code.co_freevars, code.co_cellvars,
            code.co_flags, code.co_argcount, code.co_posonlyargcount,
            code.co_kwonlyargcount, code.co_exceptiontable,
        )

    for name in ("__init__", "ask_with_char_rrf", "build_context", "close"):
        require(
            signature(getattr(pipeline_class, name).__code__)
            == signature(method_codes[name]),
            f"C199_LOADED_METHOD_MISMATCH:{name}",
        )

    parser = globals().get("dotenv_values")
    url_module = globals().get("urllib")
    require(
        callable(parser) and url_module is not None,
        "C199_C001_TRANSPORT_BINDINGS_REQUIRED",
    )
    config = parser(root / ".env", interpolate=False)
    base = str(os.environ.get("ARVAN_GATEWAY_URL") or config.get("ARVAN_GATEWAY_URL") or "").strip()
    key = str(os.environ.get("ARVAN_GATEWAY_API_KEY") or config.get("ARVAN_GATEWAY_API_KEY") or "").strip()
    url = url_module.parse.urlsplit(base)
    require(
        url.scheme == "https" and bool(url.hostname) and bool(key)
        and url.username is None and url.password is None
        and not url.query and not url.fragment,
        "C199_GATEWAY_CONFIGURATION_INVALID",
    )
    route = url.path.rstrip("/") or "/v1"
    if not route.endswith("/chat/completions"):
        route += "/chat/completions"
    endpoint = url._replace(path=route).geturl()

    class NoRedirect(url_module.request.HTTPRedirectHandler):
        def redirect_request(self, req, fp, code, msg, headers, newurl):
            return None

    opener = url_module.request.build_opener(NoRedirect())
    c148_sha = "8352cfa49bedc99c50335f57a66051a3613bdb1cf3618200dff771a8c1d654f4"
    dense = report("C148.R11_target_test_hybrid100_probe", c148_sha)

    results = []
    blockers = []
    new_calls = 0
    pipeline = pipeline_class(
        vectorstore_path=project_path(dense["dense_storage_path"]),
        collection_name=dense["dense_collection"],
        use_cache=False,
    )

    try:
        for job, job_pin, prepared, rank in jobs:
            candidate = job["candidate_id"]
            request_sha = job["request_payload_sha256"]
            completed_path = folder / f"{request_sha}.completed.json"
            started_path = folder / f"{request_sha}.started.json"
            response_path = folder / f"{request_sha}.response.json"
            input_ref = None
            output_ref = None
            answer_state = {}

            try:
                if completed_path.exists():
                    saved = json.loads(completed_path.read_bytes())
                    require(
                        saved["job_receipt"] == job_pin
                        and saved["candidate_id"] == candidate,
                        "C199_COMPLETED_JOB_MISMATCH",
                    )
                    recorded_input = load_ref(saved["input_receipt"])
                    recorded_output = load_ref(saved["output_receipt"])
                    recorded_response = load_ref(saved["response_receipt"])
                    require(
                        recorded_input["runtime_identity"]["invocation_id"]
                        == saved["invocation_id"]
                        and recorded_output["input_receipt_file_bytes_sha256"]
                        == saved["input_receipt"]["file_bytes_sha256"]
                        and recorded_output["answer_text_utf8_sha256"]
                        == text_sha(saved["answer"])
                        and recorded_response["request_payload_sha256"] == request_sha
                        and text_sha(recorded_response["response_body"])
                        == recorded_response["response_body_sha256"]
                        and json.loads(recorded_response["response_body"])[
                            "choices"
                        ][0]["message"]["content"] == saved["answer"],
                        "C199_COMPLETED_RECEIPT_MISMATCH",
                    )
                    results.append(saved)
                    continue

                require(
                    not started_path.exists() and not response_path.exists(),
                    "C199_PREVIOUS_ATTEMPT_REQUIRES_RECOVERY_NO_RESEND",
                )
                invocation = sha(
                    f"C199:{candidate}:{time.time_ns()}:{os.getpid()}".encode("utf-8")
                )
                chunks = [
                    {
                        "chunk_id": cid,
                        "text": corpus[cid]["content"],
                        "title": corpus[cid].get("title"),
                        "url": corpus[cid].get("url"),
                    }
                    for cid in rank["char_rrf_top10"]
                ]
                require(
                    [text_sha(chunk["text"]) for chunk in chunks]
                    == rank["char_rrf_top10_text_sha256"],
                    "C199_CHUNK_TEXT_MISMATCH",
                )
                runtime_identity = {
                    "scope": "OUTPUT_CAP_2400_PROFILE_PROBE",
                    "invocation_id": invocation,
                    "pipeline_source_sha256": pipeline_sha,
                    "rank_receipt_sha256": job["rank_receipt"]["file_bytes_sha256"],
                    "job_receipt_sha256": job_pin["file_bytes_sha256"],
                }

                def rank_sources(*, question, candidate_id, top_k):
                    require(
                        candidate_id == candidate and top_k == 10
                        and text_sha(question) == job["question_sha256"],
                        "C199_RUNTIME_QUERY_MISMATCH",
                    )
                    return {
                        "candidate_id": candidate,
                        "question_sha256": job["question_sha256"],
                        "selected_variant": "char_rrf",
                        "runtime_identity": runtime_identity,
                        "ordered_chunks": chunks,
                    }

                def write_receipt(role, payload):
                    nonlocal input_ref, output_ref
                    if role == "input":
                        require(input_ref is None, "C199_DUPLICATE_INPUT")
                        input_ref = save("C199.R11_product_input", payload)
                        return input_ref["file_bytes_sha256"]
                    require(
                        role == "output" and input_ref is not None and output_ref is None
                        and payload["input_receipt_file_bytes_sha256"]
                        == input_ref["file_bytes_sha256"]
                        and payload["answer_text_utf8_sha256"] == answer_state["answer_sha"],
                        "C199_OUTPUT_BINDING_MISMATCH",
                    )
                    output_ref = save("C199.R11_product_output", payload)
                    return output_ref["file_bytes_sha256"]

                class Generator:
                    def generate_answer(self, *, question, context):
                        nonlocal new_calls
                        require(input_ref is not None, "C199_INPUT_RECEIPT_REQUIRED")
                        recorded = load_ref(input_ref)
                        require(
                            question == prepared["question"]
                            and context == prepared["rendered_context"]
                            and recorded["candidate_id"] == candidate
                            and recorded["runtime_identity"] == runtime_identity
                            and recorded["question_sha256"] == job["question_sha256"]
                            and recorded["ordered_chunk_ids"] == rank["char_rrf_top10"]
                            and recorded["ordered_chunk_text_sha256"]
                            == rank["char_rrf_top10_text_sha256"]
                            and recorded["rendered_context_sha256"]
                            == text_sha(context) == job["rendered_context_sha256"],
                            "C199_PRE_GENERATION_BINDING_MISMATCH",
                        )
                        started = {
                            "candidate_id": candidate,
                            "invocation_id": invocation,
                            "request_payload_sha256": request_sha,
                            "job_receipt": job_pin,
                            "input_receipt": input_ref,
                        }
                        with started_path.open("xb") as stream:
                            stream.write(encode(started))
                            stream.flush()
                            os.fsync(stream.fileno())

                        outgoing = url_module.request.Request(
                            endpoint,
                            data=encode(job["request"]),
                            method="POST",
                            headers={
                                "Authorization": f"apikey {key}",
                                "Content-Type": "application/json",
                            },
                        )
                        new_calls += 1
                        try:
                            with opener.open(outgoing, timeout=180) as response:
                                status = response.status
                                body = response.read().decode("utf-8")
                        except url_module.error.HTTPError as error:
                            with error:
                                status = error.code
                                body = error.read().decode("utf-8")
                        except (OSError, ValueError):
                            raise ValueError("C199_UNKNOWN_DISPATCH_OUTCOME_NO_RESEND") from None

                        response_record = {
                            "candidate_id": candidate,
                            "invocation_id": invocation,
                            "request_payload_sha256": request_sha,
                            "input_receipt_file_bytes_sha256": input_ref["file_bytes_sha256"],
                            "started_receipt_sha256": sha(started_path.read_bytes()),
                            "http_status": status,
                            "response_body": body,
                            "response_body_sha256": text_sha(body),
                        }
                        atomic_create(response_path, encode(response_record))
                        require(status == 200, "C199_NON200_RESPONSE_PRESERVED")
                        envelope = json.loads(body)
                        choices = envelope.get("choices", [])
                        require(len(choices) == 1, "C199_EXPECTED_ONE_CHOICE")
                        choice = choices[0]
                        answer = choice.get("message", {}).get("content")
                        require(
                            envelope["model"].strip().rsplit("/", 1)[-1].lower()
                            in {"deepseek-v4-flash", "deepseek-v4-flash-0731"}
                            and choice.get("finish_reason") == "stop"
                            and isinstance(answer, str) and bool(answer.strip()),
                            "C199_RESPONSE_INCOMPLETE_OR_WRONG_MODEL",
                        )
                        answer_state.update(
                            answer_sha=text_sha(answer),
                            finish_reason=choice["finish_reason"],
                            usage_as_reported=envelope.get("usage"),
                            response_receipt={
                                "path": response_path.relative_to(root).as_posix(),
                                "file_bytes_sha256": sha(response_path.read_bytes()),
                            },
                        )
                        return answer

                pipeline.generator = Generator()
                returned = pipeline.ask_with_char_rrf(
                    prepared["question"],
                    candidate_id=candidate,
                    rank_sources=rank_sources,
                    write_receipt=write_receipt,
                    top_k=10,
                )
                require(
                    returned["from_cache"] is False
                    and output_ref is not None
                    and returned["input_receipt_file_bytes_sha256"]
                    == input_ref["file_bytes_sha256"]
                    and returned["output_receipt_file_bytes_sha256"]
                    == output_ref["file_bytes_sha256"],
                    "C199_PRODUCT_RETURN_MISMATCH",
                )
                saved = {
                    "candidate_id": candidate,
                    "scenario_ids": [
                        spec["regression_spec_id"] for spec in job["scenario_specs"]
                    ],
                    "invocation_id": invocation,
                    "job_receipt": job_pin,
                    "request_payload_sha256": request_sha,
                    "input_receipt": input_ref,
                    "output_receipt": output_ref,
                    "answer": returned["answer"],
                    **answer_state,
                    "semantic_verdict": None,
                }
                atomic_create(completed_path, encode(saved))
                results.append(saved)

            except Exception as error:
                blockers.append({
                    "candidate_id": candidate,
                    "type": type(error).__name__,
                    "detail": str(error),
                    "input_receipt": input_ref,
                    "response_path": (
                        response_path.relative_to(root).as_posix()
                        if response_path.exists() else None
                    ),
                })
    finally:
        pipeline.close()

    result = {
        "cell": "C199",
        "roadmap_subtask": "R11.3",
        "checklist_item": 12,
        "status": (
            "TWO_PROFILE_PROBE_RESPONSES_BOUND"
            if len(results) == 2 and not blockers else "OUTPUT_CAP_PROBE_PARTIAL"
        ),
        "input_file_hashes": {"C198": c198_sha, "C193": c193_sha, "C191": plan_sha},
        "probe_plan": probe_plan_pin,
        "profile_change": {"max_tokens": {"before": 600, "after": 2400}},
        "completed_candidate_count": len(results),
        "candidate_results": results,
        "blocked_candidates": blockers,
        "new_api_calls": new_calls,
        "current_rate_verified": False,
        "historical_new_estimate_toman": new_estimate_units / 1_000_000,
        "historical_combined_estimate_toman": combined_estimate,
        "original_600_token_results_preserved": True,
        "new_semantic_verdicts": 0,
        "done_item_count": 9,
        "open_item_count": 3,
        "open_item_ids": [9, 10, 12],
        "selected_version_gate": "BLOCKED",
    }
    result["output"] = save("C199.R11_output_cap_probe_result", result)
    print(json.dumps(result, ensure_ascii=False, indent=2))
    return result


V3_C199 = run_v3_c199()
````

### NB22E-C200 — داوری دو پاسخ پروفایل ۲۴۰۰ توکنی

زنجیرهٔ رسیدهای دو پاسخ C199 و ثابت‌ماندن درخواست به‌جز max_tokens
بررسی می‌شوند. پاسخ آموزش مالی کودک در دامنهٔ شواهد پذیرفته می‌شود.

پاسخ نرخ بهره مؤلفهٔ عددی را بیان کرده، اما اصلاح اعداد در ادامهٔ
منبع و توضیح منبع دیگر را در نتیجه‌گیری خود لحاظ نکرده است.
این ایراد جدا ثبت می‌شود.

این رأی‌ها به پروفایل ۲۴۰۰ تعلق دارند و جای نتیجهٔ ثبت‌شدهٔ اجرای
۶۰۰ توکنی را نمی‌گیرند. این سل آفلاین است.

</div>

````python
# NB22E-C200 — Assess the two completed 2400-token profile probes.
# Verify execution bindings and retain separate results for the original profile.

def run_v3_c200():
    def require(ok, code):
        if not ok:
            raise ValueError(code)

    def encode(value):
        return json.dumps(
            value, ensure_ascii=False, sort_keys=True,
            separators=(",", ":"), allow_nan=False,
        ).encode("utf-8")

    def sha(raw):
        return hashlib.sha256(raw).hexdigest()

    reports = Path("reports/phase2/quality_improvement/NB22E.lab.v3")
    cwd = Path.cwd()
    root = next(
        (p.resolve() for p in (cwd, *cwd.parents) if (p / reports).is_dir()),
        None,
    )
    require(root is not None, "C200_PROJECT_ROOT_NOT_FOUND")

    def load(relative, digest):
        path = (root / relative).resolve()
        require(path.is_relative_to(root), "C200_PATH_OUTSIDE_PROJECT")
        raw = path.read_bytes()
        require(sha(raw) == digest, f"C200_HASH_MISMATCH:{relative}")
        return json.loads(raw)

    def load_ref(pin):
        return load(pin["path"], pin["file_bytes_sha256"])

    c199_sha = "118b4d4ecf697feafcf604e13871fd44cdb845f19624e5fe3f7ac07089d51628"
    packet_sha = "0b20b53e6812c69c47809a027f2c33c59e285bc6baf509db91d4e2707f460a4b"
    execution = load(
        reports / f"C199.R11_output_cap_probe_result.{c199_sha}.json", c199_sha
    )
    packet = load(
        reports / f"C193.R11_batch_semantic_review_packet.{packet_sha}.json",
        packet_sha,
    )
    probe_plan = load_ref(execution["probe_plan"])
    require(
        execution["status"] == "TWO_PROFILE_PROBE_RESPONSES_BOUND"
        and execution["completed_candidate_count"] == 2
        and not execution["blocked_candidates"],
        "C200_SUCCESSFUL_C199_REQUIRED",
    )
    contexts = {
        row["candidate_id"]: row for row in packet["candidate_reviews"]
    }

    def witness(context, number, quote):
        matches = [
            match for match in re.finditer(
                r"\[منبع (\d+)\]\n(.*?)(?=\n\[منبع \d+\]\n|\Z)",
                context, flags=re.S,
            )
            if int(match.group(1)) == number
        ]
        require(len(matches) == 1, "C200_SOURCE_NOT_UNIQUE")
        match = matches[0]
        require(match.group(2).count(quote) == 1, "C200_EXACT_QUOTE_MISMATCH")
        chunk = re.search(r"شناسه چانک: ([^\n]+)", match.group(2))
        require(chunk is not None, "C200_CHUNK_ID_MISSING")
        start = context.index(quote, match.start(), match.end())
        return {
            "source_number": number,
            "chunk_id": chunk.group(1).strip(),
            "exact_quote": quote,
            "offset_reference": "C193_EXACT_DISPATCHED_CONTEXT",
            "span_unit": "UNICODE_CODE_POINT",
            "start": start,
            "end": start + len(quote),
        }

    child = "golden_v2_candidate_e5ea19069f2d84ca861c2847"
    interest = "golden_v2_candidate_b8756d70dae44e64533a6691"
    require(
        {row["candidate_id"] for row in execution["candidate_results"]}
        == {child, interest},
        "C200_TARGET_SET_MISMATCH",
    )
    reviews = []

    for row in execution["candidate_results"]:
        candidate = row["candidate_id"]
        job = load_ref(row["job_receipt"])
        original_job = load_ref(job["original_job_receipt"])
        input_record = load_ref(row["input_receipt"])
        output_record = load_ref(row["output_receipt"])
        response_record = load_ref(row["response_receipt"])
        prepared = load_ref(job["prepared_context_receipt"])
        context = contexts[candidate]["exact_dispatched_context"]
        request_sha = sha(encode(job["request"]))

        require(
            row["job_receipt"] in probe_plan["jobs"]
            and job["candidate_id"] == input_record["candidate_id"] == candidate
            and request_sha == row["request_payload_sha256"]
            == response_record["request_payload_sha256"]
            and job["request"]["max_tokens"] == 2400
            and original_job["request"]["max_tokens"] == 600
            and {k: v for k, v in job["request"].items() if k != "max_tokens"}
            == {k: v for k, v in original_job["request"].items() if k != "max_tokens"}
            and prepared["rendered_context"] == context
            and input_record["rendered_context_sha256"]
            == job["rendered_context_sha256"] == sha(context.encode("utf-8"))
            and input_record["runtime_identity"]["invocation_id"]
            == response_record["invocation_id"] == row["invocation_id"]
            and response_record["input_receipt_file_bytes_sha256"]
            == output_record["input_receipt_file_bytes_sha256"]
            == row["input_receipt"]["file_bytes_sha256"],
            "C200_PROFILE_OR_INVOCATION_BINDING_MISMATCH",
        )

        response_path = (root / row["response_receipt"]["path"]).resolve()
        started = load(
            response_path.with_name(f"{request_sha}.started.json").relative_to(root),
            response_record["started_receipt_sha256"],
        )
        require(
            started["candidate_id"] == candidate
            and started["invocation_id"] == row["invocation_id"]
            and started["request_payload_sha256"] == request_sha
            and started["input_receipt"] == row["input_receipt"],
            "C200_STARTED_RECEIPT_MISMATCH",
        )

        body = response_record["response_body"]
        require(
            response_record["http_status"] == 200
            and sha(body.encode("utf-8")) == response_record["response_body_sha256"],
            "C200_RAW_RESPONSE_MISMATCH",
        )
        envelope = json.loads(body)
        choice = envelope["choices"][0]
        answer = choice["message"]["content"]
        require(
            len(envelope["choices"]) == 1
            and choice["finish_reason"] == "stop"
            and envelope["model"].strip().rsplit("/", 1)[-1].lower()
            in {"deepseek-v4-flash", "deepseek-v4-flash-0731"}
            and answer == row["answer"]
            and sha(answer.encode("utf-8")) == row["answer_sha"]
            == output_record["answer_text_utf8_sha256"],
            "C200_ANSWER_BINDING_MISMATCH",
        )

        review = {
            "candidate_id": candidate,
            "scenario_ids": row["scenario_ids"],
            "request_payload_sha256": request_sha,
            "response_text_sha256": row["answer_sha"],
            "profile": job["request"] | {"messages": "BOUND_BY_REQUEST_HASH"},
            "input_receipt": row["input_receipt"],
            "output_receipt": row["output_receipt"],
            "response_receipt": row["response_receipt"],
            "original_response_receipt": job["original_response_receipt"],
        }

        if candidate == child:
            require(
                "جبران‌پذیر" in answer and "کل زندگی" in answer,
                "C200_CHILD_SCOPE_MARKERS_MISSING",
            )
            review.update(
                verdict="PASS_SCOPED_EXPECTED_BEHAVIOR_PROFILE_2400",
                reason_fa="پاسخ، یادگیری از اشتباه را به تجربهٔ جبران‌پذیر و همراهی والدین مقید کرده است.",
                witnesses=[
                    witness(
                        context, 2,
                        "اما نه اشتباهی که کل زندگیشون رو تحت تأثیر قرار بده",
                    ),
                    witness(
                        context, 3,
                        "اما آزمایش‌هایی که یادگیری داره، جبران‌پذیره،",
                    ),
                    witness(
                        context, 7,
                        "کودک ممکن است بخشی از پولش را گم کند، اشتباه خرج کند یا خراب کند. این بخشی از یادگیری است.",
                    ),
                ],
            )
        else:
            require(
                "منفی ۱۰٪" in answer and "پاسخ قابل اتکا نیست" in answer,
                "C200_INTEREST_RESPONSE_MARKERS_MISSING",
            )
            review.update(
                verdict="NUMERIC_COMPONENT_PRESENT_SOURCE_READING_ISSUE_OPEN",
                reason_fa=(
                    "پاسخ منفی ۱۰٪ را بیان می‌کند؛ اما در استدلال امتناع، "
                    "اصلاح اعداد در ادامهٔ منبع ۱ و توضیح منبع ۲ را لحاظ نکرده است."
                ),
                witnesses=[
                    witness(
                        context, 1,
                        "نرخ تورم ۲۰ درصده. نرخ تورم ۲۰ درصده، نرخ بهره ۳۰ درصده.",
                    ),
                    witness(
                        context, 2,
                        "حتی یه پس‌انداز منفیه.",
                    ),
                ],
                full_scenario_pass_granted=False,
            )
        reviews.append(review)

    result = {
        "cell": "C200",
        "roadmap_subtask": "R11.3",
        "checklist_item": 12,
        "status": "TWO_OUTPUT_CAP_PROBE_RESPONSES_REVIEWED",
        "input_file_hashes": {"C199": c199_sha, "C193_review_packet": packet_sha},
        "profile": {"max_tokens": 2400},
        "reviews": reviews,
        "profile_2400_scoped_pass_count": 1,
        "profile_2400_content_issue_count": 1,
        "original_600_token_execution_error_count_preserved": 2,
        "main_profile_scoped_pass_count_including_savings": 8,
        "new_api_calls": 0,
        "done_item_count": 9,
        "open_item_count": 3,
        "open_item_ids": [9, 10, 12],
        "selected_version_gate": "BLOCKED",
    }

    raw = (
        json.dumps(result, ensure_ascii=False, sort_keys=True, indent=2) + "\n"
    ).encode("utf-8")
    digest = sha(raw)
    relative = reports / f"C200.R11_output_cap_probe_review.{digest}.json"
    path = root / relative
    try:
        with path.open("xb") as stream:
            stream.write(raw)
    except FileExistsError:
        pass
    require(path.read_bytes() == raw, "C200_OUTPUT_RELOAD_MISMATCH")
    result["output"] = {
        "path": relative.as_posix(),
        "file_bytes_sha256": digest,
        "reload_validation": "PASS",
    }
    print(json.dumps(result, ensure_ascii=False, indent=2))
    return result


V3_C200 = run_v3_c200()
````

### NB22E-C201 — استخراج شواهد اصلی شش مانع دامنهٔ مرجع

برای شش سناریوی مسدود، رکورد کامل Golden و مشخصات اصلی C078 با هش
فایل خوانده می‌شوند. این رکوردها کنار سؤال، زمینه و پاسخ منتخب C193
و مانع ثبت‌شده در C198 قرار می‌گیرند.

هدف، بررسی مرجع و دامنهٔ انتظار همان سناریوهاست. منشأ و محدودیت‌های
مرجع تاریخی حفظ می‌شوند. این سل آفلاین است و رأی تازه‌ای صادر نمی‌کند.

</div>

````python
# NB22E-C201 — Extract the original records for six reference-scope blockers.
# Bind Golden rows and original specifications to the selected response evidence.

def run_v3_c201():
    def require(ok, code):
        if not ok:
            raise ValueError(code)

    def sha(raw):
        return hashlib.sha256(raw).hexdigest()

    reports = Path("reports/phase2/quality_improvement/NB22E.lab.v3")
    cwd = Path.cwd()
    root = next(
        (path.resolve() for path in (cwd, *cwd.parents) if (path / reports).is_dir()),
        None,
    )
    require(root is not None, "C201_PROJECT_ROOT_NOT_FOUND")

    def read_bound(relative, digest):
        path = (root / relative).resolve()
        require(path.is_relative_to(root), "C201_PATH_OUTSIDE_PROJECT")
        raw = path.read_bytes()
        require(sha(raw) == digest, f"C201_HASH_MISMATCH:{relative}")
        return raw

    def load_report(stem, digest):
        return json.loads(read_bound(reports / f"{stem}.{digest}.json", digest))

    ledger_sha = "e9c605ecf404f084c0953b00f01e5d8dcb7752885aa720ffa83bf19c7789e6a3"
    packet_sha = "0b20b53e6812c69c47809a027f2c33c59e285bc6baf509db91d4e2707f460a4b"
    golden_sha = "b9ca18e710686ddca48a25c637840b716adb306717cafd8ec7c96ce9e182e6f2"
    specs_sha = "09b05f51fbadd8a090b14c1c2ee65036987aa8e36555354535b983ebffe02006"

    ledger = load_report("C198.R11_selected_citation_witness_disposition", ledger_sha)
    packet = load_report("C193.R11_batch_semantic_review_packet", packet_sha)
    blocked = [
        row for row in ledger["scenario_dispositions"]
        if row["scenario_disposition"] == "BLOCKED_REFERENCE_SCOPE_NOT_ESTABLISHED"
    ]
    expected_ids = {
        "REG_c05a077fb1b2f216431f2cde",
        "REG_e7a80af5ef7981de0218a370",
        "REG_774463999d5a4672e41159e2",
        "REG_c7935714c21dfa4cf5e7f114",
        "REG_14116eb18411af97e360f237",
        "REG_0832dd8d154ea625634caad9",
    }
    require(
        len(blocked) == 6
        and {row["regression_spec_id"] for row in blocked} == expected_ids,
        "C201_REFERENCE_BLOCKER_SET_MISMATCH",
    )

    candidate_ids = {row["candidate_id"] for row in blocked}
    selected = {
        row["candidate_id"]: row for row in packet["candidate_reviews"]
        if row["candidate_id"] in candidate_ids
    }
    require(
        len(selected) == len(candidate_ids) == 6,
        "C201_SELECTED_CANDIDATE_SET_MISMATCH",
    )

    golden_path = "artifacts/phase2/golden_dataset_v2/golden_dataset_v2.jsonl"
    golden_rows = {}
    for number, line in enumerate(
        read_bound(golden_path, golden_sha).splitlines(), 1
    ):
        if not line.strip():
            continue
        record = json.loads(line)
        candidate = record.get("candidate_id")
        if candidate in candidate_ids:
            require(candidate not in golden_rows, "C201_DUPLICATE_GOLDEN_ROW")
            golden_rows[candidate] = {
                "record": record,
                "source": {
                    "path": golden_path,
                    "file_bytes_sha256": golden_sha,
                    "line_number": number,
                    "row_sha256_excluding_line_ending": sha(line),
                },
            }
    require(
        golden_rows.keys() == candidate_ids,
        "C201_GOLDEN_CANDIDATE_MISSING",
    )

    specs_path = (
        "reports/phase2/evaluation/NB21C.C078.action42_final/"
        f"C078.regression_specs.{specs_sha}.json"
    )
    original_specs = json.loads(read_bound(specs_path, specs_sha))
    require(isinstance(original_specs, list), "C201_ORIGINAL_SPEC_ARRAY_REQUIRED")
    spec_index = {}
    for index, spec in enumerate(original_specs):
        if spec.get("regression_spec_id") in expected_ids:
            spec_id = spec["regression_spec_id"]
            require(spec_id not in spec_index, "C201_DUPLICATE_ORIGINAL_SPEC")
            spec_index[spec_id] = {
                "record": spec,
                "source": {
                    "path": specs_path,
                    "file_bytes_sha256": specs_sha,
                    "json_pointer": f"/{index}",
                },
            }
    require(spec_index.keys() == expected_ids, "C201_ORIGINAL_SPEC_MISSING")

    cases = []
    for row in blocked:
        candidate = row["candidate_id"]
        spec_id = row["regression_spec_id"]
        golden = golden_rows[candidate]
        original = spec_index[spec_id]
        current = selected[candidate]
        question = golden["record"]["evaluation_content"]["question"]
        context = current["exact_dispatched_context"]
        answer = current["answer"]
        require(
            original["record"]["candidate_id"] == candidate
            and question == current["question"]
            and sha(question.encode("utf-8")) == row["question_sha256"]
            and sha(context.encode("utf-8")) == row["rendered_context_sha256"]
            and sha(answer.encode("utf-8")) == row["response_text_sha256"],
            "C201_QUESTION_OR_RESPONSE_BINDING_MISMATCH",
        )
        cases.append({
            "candidate_id": candidate,
            "regression_spec_id": spec_id,
            "question": question,
            "required_reference_decision_fa": row["required_reference_decision_fa"],
            "golden_source": golden["source"],
            "original_golden_record": golden["record"],
            "original_spec_source": original["source"],
            "original_regression_spec": original["record"],
            "selected_response_text_sha256": row["response_text_sha256"],
            "selected_response": answer,
            "selected_context_sha256": row["rendered_context_sha256"],
            "exact_selected_context": context,
            "reference_scope_verdict": None,
        })

    result = {
        "cell": "C201",
        "roadmap_subtask": "R11.3",
        "checklist_item": 12,
        "status": "SIX_ORIGINAL_REFERENCE_CASES_BOUND_FOR_REVIEW",
        "input_file_hashes": {
            "C198": ledger_sha,
            "C193_review_packet": packet_sha,
            "golden": golden_sha,
            "C078_original_specs": specs_sha,
        },
        "case_count": 6,
        "reference_cases": cases,
        "new_reference_judgments": 0,
        "new_api_calls": 0,
        "done_item_count": 9,
        "open_item_count": 3,
        "open_item_ids": [9, 10, 12],
        "selected_version_gate": "BLOCKED",
    }

    raw = (
        json.dumps(result, ensure_ascii=False, sort_keys=True, indent=2) + "\n"
    ).encode("utf-8")
    digest = sha(raw)
    relative = reports / f"C201.R11_original_reference_review_packet.{digest}.json"
    path = root / relative
    try:
        with path.open("xb") as stream:
            stream.write(raw)
    except FileExistsError:
        pass
    require(path.read_bytes() == raw, "C201_OUTPUT_RELOAD_MISMATCH")

    summary = {
        key: value for key, value in result.items() if key != "reference_cases"
    }
    summary["questions"] = [case["question"] for case in cases]
    summary["output"] = {
        "path": relative.as_posix(),
        "file_bytes_sha256": digest,
        "reload_validation": "PASS",
    }
    print(json.dumps(summary, ensure_ascii=False, indent=2))
    return summary


V3_C201 = run_v3_c201()
````

````python
# NB22E-C209 — Complete only the 65 authorized local fixed-pool jobs.
# C001 imports:
# Path, json, hashlib, os, tempfile, ast, collections,
# math, re, unicodedata, time.

def run_v3_c209():
    required = (
        "Path", "json", "hashlib", "os", "tempfile", "ast",
        "collections", "math", "re", "unicodedata", "time",
    )
    missing_imports = [name for name in required if name not in globals()]
    if missing_imports:
        raise RuntimeError(
            "C209_IMPORTS_REQUIRED_IN_C001:" + ",".join(missing_imports)
        )

    started = time.perf_counter()

    def require(ok, code):
        if not ok:
            raise ValueError(code)

    def sha(raw):
        return hashlib.sha256(raw).hexdigest()

    def canonical(value):
        return json.dumps(
            value, ensure_ascii=False, sort_keys=True,
            separators=(",", ":"), allow_nan=False,
        ).encode("utf-8")

    def encoded(value):
        return (
            json.dumps(
                value, ensure_ascii=False, sort_keys=True,
                indent=2, allow_nan=False,
            ) + "\n"
        ).encode("utf-8")

    root = Path(
        "/Users/macbookpro/AMIR_DATA/00_WORKS/01_PROJECTS/"
        "RAG-projects/RAG_finance"
    ).resolve()
    folder = root / "reports/phase2/quality_improvement/NB22E.lab.v3"
    inputs = {}

    def atomic_write(path, raw):
        if path.exists():
            require(path.read_bytes() == raw, "C209_OUTPUT_CONFLICT")
        else:
            fd, temporary = tempfile.mkstemp(prefix=".C209-", dir=str(path.parent))
            try:
                with os.fdopen(fd, "wb") as stream:
                    stream.write(raw)
                    stream.flush()
                    os.fsync(stream.fileno())
                try:
                    os.link(temporary, path)
                except FileExistsError:
                    require(path.read_bytes() == raw, "C209_OUTPUT_CONFLICT")
            finally:
                if os.path.exists(temporary):
                    os.unlink(temporary)
        require(path.read_bytes() == raw, "C209_RELOAD_CONFLICT")
        return {
            "path": path.relative_to(root).as_posix(),
            "file_bytes_sha256": sha(raw),
            "reload_validation": "PASS",
        }

    def save(stem, value):
        raw = encoded(value)
        return atomic_write(folder / (stem + "." + sha(raw) + ".json"), raw)

    def read_bytes(relative, digest):
        path = (root / relative).resolve()
        require(path.is_relative_to(root), "C209_PATH_OUTSIDE_PROJECT")
        raw = path.read_bytes()
        require(sha(raw) == digest, "C209_HASH_MISMATCH:" + relative)
        inputs[relative] = digest
        return raw

    def read_bound(relative, digest, jsonl=False):
        raw = read_bytes(relative, digest)
        if jsonl:
            return [json.loads(line) for line in raw.splitlines() if line.strip()]
        return json.loads(raw)

    def report(stem, digest):
        return read_bound(
            folder.relative_to(root).as_posix()
            + "/" + stem + "." + digest + ".json",
            digest,
        )

    manifest = report(
        "C208.R11_fixed_pool_missing_work",
        "55fe3293cf1098183dda3c9c7a276d35922bb5d9867aa894bbadffebe521be86",
    )
    previous = report(
        "C207.R11_independent_fixed_pool_arm_results",
        "e553f9f083ae6820eccf06cc2c69cdb017122637d38c64692448f11fade8726a",
    )
    frozen = report(
        "C024.R10_selected_evidence_freeze",
        "b807237f302b824c5ad943b5e3adb68a79ebd5c29c51c30edc488b88deec82e8",
    )
    jobs = manifest["jobs"]
    require(
        len(jobs) == len({job["candidate_id"] for job in jobs}) == 65
        and sum(len(job["scenario_ids"]) for job in jobs) == 98,
        "C209_AUTHORIZED_JOB_POPULATION_MISMATCH",
    )

    arm_plan_pin = previous["plan_receipt"]
    arm_plan = read_bound(
        arm_plan_pin["path"], arm_plan_pin["file_bytes_sha256"]
    )
    config = arm_plan["configuration"]
    code_pin = previous["code_receipt"]
    arm_source = read_bytes(
        code_pin["path"], code_pin["file_bytes_sha256"]
    ).decode("utf-8")
    require(config["arm_id"] == "pool50_char_rrf_v1", "C209_ARM_MISMATCH")
    arm_namespace = {}
    exec(compile(arm_source, code_pin["path"], "exec"), arm_namespace)
    rank_pool50 = arm_namespace["rank_pool50"]

    selected = frozen["selected_method"]
    producer_cells = selected["producer_code_snapshot"]
    lexical_config = selected["configuration"]
    require(
        sha(canonical(producer_cells)) == selected["producer_code_sha256"]
        and sha(canonical(lexical_config))
        == selected["configuration_canonical_sha256"],
        "C209_FROZEN_PRODUCER_IDENTITY_MISMATCH",
    )

    # Load only the frozen lexical functions, not notebook execution/history.
    wanted = {"normalized_words", "char_tokens", "build_index", "lexical_search"}
    functions = {}
    for source in producer_cells:
        for node in ast.parse(source).body:
            if isinstance(node, ast.FunctionDef) and node.name in wanted:
                require(
                    node.name not in functions and not node.decorator_list,
                    "C209_FROZEN_FUNCTION_CONFLICT:" + node.name,
                )
                functions[node.name] = node
    require(set(functions) == wanted, "C209_FROZEN_FUNCTIONS_MISSING")
    lexical_namespace = {
        "CONFIG": lexical_config,
        "collections": collections,
        "math": math,
        "re": re,
        "unicodedata": unicodedata,
        "require": require,
    }
    module = ast.Module(body=list(functions.values()), type_ignores=[])
    exec(
        compile(module, "<C209:pinned-lexical-functions>", "exec"),
        lexical_namespace,
    )

    bindings = frozen["evaluation_lock"]["data_bindings"]
    corpus_rows = read_bound(
        bindings["chunks"]["path"],
        bindings["chunks"]["file_bytes_sha256"],
        jsonl=True,
    )
    golden_rows = read_bound(
        bindings["golden"]["path"],
        bindings["golden"]["file_bytes_sha256"],
        jsonl=True,
    )
    chunks = {row["chunk_id"]: row for row in corpus_rows}
    require(len(chunks) == len(corpus_rows), "C209_DUPLICATE_CORPUS_CHUNK")

    wanted_ids = {job["candidate_id"] for job in jobs}
    questions = {}
    for row in golden_rows:
        cid = row["candidate_id"]
        if cid in wanted_ids:
            require(cid not in questions, "C209_DUPLICATE_GOLDEN_QUERY")
            questions[cid] = row["evaluation_content"]["question"]
    require(set(questions) == wanted_ids, "C209_QUESTION_COVERAGE_MISMATCH")

    for job in jobs:
        cid = job["candidate_id"]
        require(
            sha(questions[cid].encode("utf-8")) == job["question_sha256"]
            and len(job["fixed_pool50_and_baseline_order"]) == 50
            and len(set(job["fixed_pool50_and_baseline_order"])) == 50
            and set(job["fixed_pool50_and_baseline_order"]).issubset(chunks)
            and set(job["reference_chunk_ids"]).issubset(chunks),
            "C209_QUERY_POOL_OR_REFERENCE_MISMATCH:" + cid,
        )

    run_plan = {
        "arm_id": config["arm_id"],
        "authorization_scope": "65_MISSING_LOCAL_CHAR_AND_FIXED_POOL_JOBS",
        "arm_code_receipt": code_pin,
        "arm_configuration": config,
        "lexical_producer_sha256": selected["producer_code_sha256"],
        "lexical_configuration_sha256":
            selected["configuration_canonical_sha256"],
        "question_preprocessing": "STRIP_BEFORE_LEXICAL_SEARCH",
        "source_context": "SAME_FROZEN_CORPUS_AND_LEXICAL_CONFIGURATION_AS_C189",
        "jobs": jobs,
        "input_file_hashes": dict(inputs),
        "new_api_calls": 0,
        "new_model_pairs": 0,
    }
    plan_pin = save("C209.R11_fixed_pool_local_completion_plan", run_plan)
    checkpoint_folder = (
        folder / "C209.R11_fixed_pool_checkpoints"
        / plan_pin["file_bytes_sha256"]
    )
    checkpoint_folder.mkdir(parents=True, exist_ok=True)

    def measure(order, gold):
        flags = [int(chunk_id in gold) for chunk_id in order[:10]]
        ideal = math.fsum(
            1 / math.log2(i + 2) for i in range(min(10, len(gold)))
        )
        return {
            "hit_at_10": int(any(flags)),
            "recall_at_10": sum(flags) / len(gold),
            "ndcg_at_10": math.fsum(
                flag / math.log2(i + 2) for i, flag in enumerate(flags)
            ) / ideal,
        }

    index = None
    new_char_runs = 0
    reused = 0
    completed_rows = []
    scenario_rows = []
    blockers = []

    for number, job in enumerate(jobs, 1):
        cid = job["candidate_id"]
        checkpoint = checkpoint_folder / (sha(canonical(job)) + ".json")
        try:
            if checkpoint.exists():
                record = json.loads(checkpoint.read_bytes())
                payload = record["payload"]
                require(
                    record["payload_sha256"] == sha(canonical(payload))
                    and payload["job"] == job
                    and payload["plan_sha256"] == plan_pin["file_bytes_sha256"],
                    "C209_CHECKPOINT_IDENTITY_MISMATCH",
                )
                reused += 1
            else:
                if index is None:
                    print("C209: building the frozen local character index", flush=True)
                    index = lexical_namespace["build_index"](chunks, "char")

                char_order = lexical_namespace["lexical_search"](
                    index, questions[cid].strip(), "char"
                )
                new_char_runs += 1
                require(
                    isinstance(char_order, list)
                    and len(char_order) == len(set(char_order))
                    and set(char_order).issubset(chunks),
                    "C209_INVALID_CHAR_ORDER",
                )
                baseline = job["fixed_pool50_and_baseline_order"]
                order, scores = rank_pool50(
                    list(baseline), baseline, char_order, config
                )
                gold = set(job["reference_chunk_ids"])
                payload = {
                    "job": job,
                    "plan_sha256": plan_pin["file_bytes_sha256"],
                    "char_channel_order": char_order,
                    "candidate_order": order,
                    "scores": scores,
                    "baseline_metrics": measure(baseline, gold),
                    "candidate_metrics": measure(order, gold),
                }
                atomic_write(checkpoint, encoded({
                    "payload": payload,
                    "payload_sha256": sha(canonical(payload)),
                }))

            baseline = job["fixed_pool50_and_baseline_order"]
            order, scores = rank_pool50(
                list(baseline), baseline, payload["char_channel_order"], config
            )
            gold = set(job["reference_chunk_ids"])
            require(
                order == payload["candidate_order"]
                and scores == payload["scores"]
                and measure(baseline, gold) == payload["baseline_metrics"]
                and measure(order, gold) == payload["candidate_metrics"],
                "C209_CHECKPOINT_RESULT_MISMATCH",
            )
            completed_rows.append({
                "candidate_id": cid,
                "scenario_ids": job["scenario_ids"],
                "baseline_metrics": payload["baseline_metrics"],
                "candidate_metrics": payload["candidate_metrics"],
                "checkpoint": {
                    "path": checkpoint.relative_to(root).as_posix(),
                    "file_bytes_sha256": sha(checkpoint.read_bytes()),
                },
            })
            for spec_id in job["scenario_ids"]:
                scenario_rows.append({
                    "regression_spec_id": spec_id,
                    "candidate_id": cid,
                    "status": (
                        "SCOPED_REPAIR_TARGET_MET"
                        if payload["candidate_metrics"]["hit_at_10"]
                        else "SCOPED_REPAIR_TARGET_NOT_MET"
                    ),
                    "baseline_hit_at_10":
                        payload["baseline_metrics"]["hit_at_10"],
                    "candidate_hit_at_10":
                        payload["candidate_metrics"]["hit_at_10"],
                    "aggregate_acceptance_granted": False,
                })
            print(
                f"C209 {number}/65 complete | "
                f"new_char_runs={new_char_runs} | reused={reused}",
                flush=True,
            )
        except Exception as exc:
            blockers.append({
                "candidate_id": cid,
                "scenario_ids": job["scenario_ids"],
                "type": type(exc).__name__,
                "detail": str(exc),
            })
            for spec_id in job["scenario_ids"]:
                scenario_rows.append({
                    "regression_spec_id": spec_id,
                    "candidate_id": cid,
                    "status": "BLOCKED_LOCAL_EXECUTION",
                    "reason": str(exc),
                })
            print(f"C209 blocked: {cid} | {type(exc).__name__}", flush=True)

    preserved_queries = [
        row for row in previous["query_results"]
        if row["status"] == "OFFLINE_FIXED_POOL_ARM_EVALUATED"
    ]
    preserved_ids = {row["candidate_id"] for row in preserved_queries}
    require(
        len(preserved_queries) == 5
        and not (preserved_ids & wanted_ids),
        "C209_PRESERVED_QUERY_SET_MISMATCH",
    )
    preserved_scenarios = [
        row for row in previous["scenario_results"]
        if row["candidate_id"] in preserved_ids
    ]
    all_scenarios = preserved_scenarios + scenario_rows
    all_measured = preserved_queries + completed_rows
    require(
        len(all_scenarios)
        == len({row["regression_spec_id"] for row in all_scenarios}) == 104,
        "C209_FINAL_SCENARIO_ACCOUNTING_MISMATCH",
    )

    paired = {}
    for metric in ("hit_at_10", "recall_at_10", "ndcg_at_10"):
        before = math.fsum(
            row["baseline_metrics"][metric] for row in all_measured
        ) / len(all_measured)
        after = math.fsum(
            row["candidate_metrics"][metric] for row in all_measured
        ) / len(all_measured)
        paired[metric] = {
            "baseline": before,
            "candidate": after,
            "candidate_minus_baseline": after - before,
        }

    counts = {}
    for row in all_scenarios:
        status = row["status"]
        counts[status] = counts.get(status, 0) + 1

    result = {
        "cell": "C209",
        "roadmap_subtask": "R11.3",
        "checklist_item": 10,
        "status": (
            "FIXED_POOL_LOCAL_COMPLETION_PARTIAL"
            if blockers else "ALL_70_FIXED_POOL_QUERIES_EVALUATED"
        ),
        "arm_id": config["arm_id"],
        "plan_receipt": plan_pin,
        "input_file_hashes": inputs,
        "preserved_c207_query_count": 5,
        "completed_missing_query_count": len(completed_rows),
        "reused_missing_query_checkpoints": reused,
        "total_evaluated_query_count": len(all_measured),
        "scenario_count": 104,
        "scenario_status_counts": counts,
        "scenario_results": all_scenarios,
        "new_query_results": completed_rows,
        "blocked_queries": blockers,
        "paired_descriptive_comparison": {
            "unique_query_count": len(all_measured),
            "full_70_query_coverage": len(all_measured) == 70,
            "metrics": paired,
            "formal_nonregression_pass": False,
        },
        "formal_nonregression_status": "INFERENCE_NOT_ESTABLISHED_NOT_PASS",
        "remaining_aggregate_requirement": (
            "BIND_CONTRACT_GROUP_MEMBERSHIP_AND_COMPUTE_VALID_PAIRED_INTERVAL"
        ),
        "item10_resolved_part": (
            "INDEPENDENT_FIXED_POOL_OFFLINE_CASE_RESULTS_RECORDED"
        ),
        "selected_version_condition_still_open": (
            "NO_TRANSFER_TO_EXPANDED_SELECTED_VERSION_OR_156_SCENARIO_GROUP"
        ),
        "new_api_calls": 0,
        "new_model_pairs": 0,
        "new_lexical_scoring_runs": new_char_runs,
        "elapsed_local_seconds": time.perf_counter() - started,
        "done_item_count": 9,
        "open_item_count": 3,
        "open_item_ids": [9, 10, 12],
        "selected_version_gate": "BLOCKED",
    }
    output = save("C209.R11_fixed_pool_local_completion", result)
    summary = {
        key: value for key, value in result.items()
        if key not in (
            "input_file_hashes", "scenario_results", "new_query_results",
        )
    }
    summary["output"] = output
    print(json.dumps(summary, ensure_ascii=False, indent=2))
    return summary


V3_C209 = run_v3_c209()
````

````python
# NB22E-C210 — Close the available group-source check without inventing a CI.
# C001 imports: Path, json, hashlib, os, tempfile.

def run_v3_c210():
    def require(ok, code):
        if not ok:
            raise ValueError(code)

    def sha(raw):
        return hashlib.sha256(raw).hexdigest()

    root = Path(
        "/Users/macbookpro/AMIR_DATA/00_WORKS/01_PROJECTS/"
        "RAG-projects/RAG_finance"
    ).resolve()
    folder = root / "reports/phase2/quality_improvement/NB22E.lab.v3"
    input_receipts = {}

    def read_bound(relative, digest):
        path = (root / relative).resolve()
        require(path.is_relative_to(root), "C210_PATH_OUTSIDE_PROJECT")
        raw = path.read_bytes()
        require(sha(raw) == digest, "C210_HASH_MISMATCH:" + relative)
        input_receipts[relative] = digest
        return json.loads(raw)

    result_sha = (
        "9cd518b16952474a2988ce7bdab4d291ec6b1aa98d843a14399d74e45cf52860"
    )
    execution = read_bound(
        folder.relative_to(root).as_posix()
        + "/C209.R11_fixed_pool_local_completion."
        + result_sha + ".json",
        result_sha,
    )

    plan_sha = (
        "f387c618ceb487c24ec931e9526acf1d40108509b1f40a50343d415836e2c3ed"
    )
    comparison = read_bound(
        "artifacts/phase2/evaluation_and_ablation/work/outputs/"
        "C032.comparison_and_ablation_plan." + plan_sha + ".json",
        plan_sha,
    )
    group_pin = comparison["input_bindings"]["development_groups"]
    grouping = read_bound(
        group_pin["path"], group_pin["file_bytes_sha256"]
    )

    contract_sha = (
        "edc54bb349c632e3424aceb8da881d124d681786bf0f384be0dffdfca0ae2a6b"
    )
    contract = read_bound(
        "reports/phase2/evaluation/NB21C.C025.action12_contract/"
        "C025.execution_contract." + contract_sha + ".json",
        contract_sha,
    )

    require(
        execution["status"] == "ALL_70_FIXED_POOL_QUERIES_EVALUATED"
        and execution["total_evaluated_query_count"] == 70
        and execution["scenario_count"] == 104
        and not execution["blocked_queries"],
        "C210_EXECUTION_NOT_COMPLETE",
    )

    scenarios = execution["scenario_results"]
    query_ids = {row["candidate_id"] for row in scenarios}
    scenario_ids = {row["regression_spec_id"] for row in scenarios}
    require(
        len(query_ids) == 70
        and len(scenarios) == len(scenario_ids) == 104,
        "C210_SCENARIO_ACCOUNTING_MISMATCH",
    )
    counts = {}
    for row in scenarios:
        status = row["status"]
        counts[status] = counts.get(status, 0) + 1
    require(
        counts == {
            "SCOPED_REPAIR_TARGET_MET": 31,
            "SCOPED_REPAIR_TARGET_NOT_MET": 73,
        },
        "C210_RESULT_COUNTS_CHANGED",
    )

    group_by_query = {}
    for group in grouping["groups"]:
        members = group["candidate_ids"]
        require(
            group["query_count"] == len(members),
            "C210_GROUP_SIZE_MISMATCH",
        )
        for cid in members:
            require(cid not in group_by_query, "C210_DUPLICATE_GROUP_MEMBER")
            group_by_query[cid] = group["nb21_question_group_id"]

    require(
        grouping["split"] == "development"
        and len(group_by_query) == grouping["development_query_count"] == 2664,
        "C210_GROUP_SOURCE_SCOPE_CHANGED",
    )
    covered = sorted(query_ids & set(group_by_query))
    uncovered = sorted(query_ids - set(group_by_query))
    require(
        not covered and len(uncovered) == 70,
        "C210_GROUP_COVERAGE_CHANGED_REVIEW_REQUIRED",
    )

    final_uncertainty = contract["uncertainty_policy"]
    result = {
        "cell": "C210",
        "roadmap_subtask": "R11.3",
        "checklist_item": 10,
        "status": "FIXED_POOL_CASE_EXECUTION_COMPLETE_AGGREGATE_INFERENCE_BLOCKED",
        "arm_id": execution["arm_id"],
        "input_file_hashes": input_receipts,
        "case_execution": {
            "unique_queries": 70,
            "scenarios": 104,
            "scenario_status_counts": counts,
            "results_receipt_sha256": result_sha,
            "reexecution_required": False,
        },
        "paired_descriptive_comparison_preserved":
            execution["paired_descriptive_comparison"],
        "group_source_check": {
            "source_split": grouping["split"],
            "source_query_count": len(group_by_query),
            "covered_test_queries": len(covered),
            "uncovered_test_queries": uncovered,
            "status": "DEVELOPMENT_GROUP_SOURCE_DOES_NOT_COVER_TARGET_TEST",
            "absence_claim_scope": "BOUND_C032_TO_C030_CHAIN_ONLY",
        },
        "aggregate_inference": {
            "status": "BLOCKED_NO_COMPATIBLE_TEST_GROUP_MEMBERSHIP",
            "confidence_interval": None,
            "formal_nonregression_pass": False,
            "independent_query_groups_assumed": False,
            "new_grouping_created": False,
            "governing_contract": "NB21C.C025",
            "governing_uncertainty_policy": final_uncertainty,
            "draft_C032_seed_used": False,
            "required_evidence": (
                "HASH_BOUND_DEPENDENCY_GROUP_MEMBERSHIP_FOR_THE_70_TEST_QUERIES;"
                "COMPATIBLE_WITH_THE_CONTRACT_RESAMPLING_DESIGN"
            ),
            "status_change_condition": (
                "VALID_GROUP_MEMBERSHIP_BOUND_AND_PAIRED_INFERENCE_COMPLETED"
            ),
        },
        "item10_disposition": {
            "completed_part": (
                "INDEPENDENT_POOL50_ARM_REGISTERED_AND_ALL_CASES_EVALUATED"
            ),
            "repair_targets_unmet": 73,
            "aggregate_requirement": "BLOCKED_WITH_EXPLICIT_CAUSE",
            "whole_item_complete": False,
        },
        "scope_limits": {
            "expanded_selected_version_acceptance_granted": False,
            "declared_pool_156_acceptance_granted": False,
            "phase2_acceptance_granted": False,
            "historical_results_overwritten": False,
        },
        "new_api_calls": 0,
        "new_model_runs": 0,
        "new_ranking_runs": 0,
        "bootstrap_runs": 0,
        "done_item_count": 9,
        "open_item_count": 3,
        "open_item_ids": [9, 10, 12],
        "newly_completed_items_from_previous_open_set": [],
        "selected_version_gate": "BLOCKED",
        "next_action": (
            "CONTINUE_CHECKLIST_WITH_RECORDED_ITEM9_AND_ITEM10_BLOCKERS;"
            "NO_GENERAL_BASELINE_OR_GROUP_SEARCH"
        ),
    }

    encoded = (
        json.dumps(
            result, ensure_ascii=False, sort_keys=True,
            indent=2, allow_nan=False,
        ) + "\n"
    ).encode("utf-8")
    digest = sha(encoded)
    output = folder / (
        "C210.R11_fixed_pool_aggregate_disposition." + digest + ".json"
    )

    if output.exists():
        require(output.read_bytes() == encoded, "C210_OUTPUT_CONFLICT")
    else:
        fd, temporary = tempfile.mkstemp(prefix=".C210-", dir=str(folder))
        try:
            with os.fdopen(fd, "wb") as stream:
                stream.write(encoded)
                stream.flush()
                os.fsync(stream.fileno())
            try:
                os.link(temporary, output)
            except FileExistsError:
                require(output.read_bytes() == encoded, "C210_OUTPUT_CONFLICT")
        finally:
            if os.path.exists(temporary):
                os.unlink(temporary)

    require(
        sha(output.read_bytes()) == digest
        and json.loads(output.read_bytes()) == result,
        "C210_RELOAD_VALIDATION_FAILED",
    )

    summary = {
        "cell": "C210",
        "roadmap_subtask": "R11.3",
        "status": result["status"],
        "arm_id": result["arm_id"],
        "unique_queries": 70,
        "scenario_status_counts": counts,
        "group_source_split": grouping["split"],
        "target_group_coverage": "0/70",
        "formal_nonregression_status":
            result["aggregate_inference"]["status"],
        "new_api_calls": 0,
        "new_model_runs": 0,
        "new_ranking_runs": 0,
        "bootstrap_runs": 0,
        "done_item_count": 9,
        "open_item_count": 3,
        "open_item_ids": [9, 10, 12],
        "selected_version_gate": "BLOCKED",
        "output": {
            "path": output.relative_to(root).as_posix(),
            "file_bytes_sha256": digest,
            "reload_validation": "PASS",
        },
    }
    print(json.dumps(summary, ensure_ascii=False, indent=2))
    return summary


V3_C210 = run_v3_c210()
````

````python
# NB22E-C216 — At most four authorized policy-control requests.
# One attempt per unique request; no automatic retry.
# C001 imports: Path, json, hashlib, os, tempfile, time,
# urllib.parse/request/error, dotenv_values.

def run_v3_c216():
    def require(ok, code):
        if not ok:
            raise ValueError(code)

    def sha(raw):
        return hashlib.sha256(raw).hexdigest()

    def encode(value):
        return json.dumps(
            value, ensure_ascii=False, sort_keys=True,
            separators=(",", ":"), allow_nan=False,
        ).encode("utf-8")

    root = Path(
        "/Users/macbookpro/AMIR_DATA/00_WORKS/01_PROJECTS/"
        "RAG-projects/RAG_finance"
    ).resolve()
    folder = root / "reports/phase2/quality_improvement/NB22E.lab.v3"
    inputs = {}

    def read_bytes(pin):
        path = (root / pin["path"]).resolve()
        require(path.is_relative_to(root), "C216_PATH_OUTSIDE_PROJECT")
        raw = path.read_bytes()
        require(
            sha(raw) == pin["file_bytes_sha256"],
            "C216_HASH_MISMATCH:" + pin["path"],
        )
        inputs[pin["path"]] = pin["file_bytes_sha256"]
        return raw

    def read_bound(pin):
        return json.loads(read_bytes(pin))

    def write_once(path, raw):
        path.parent.mkdir(parents=True, exist_ok=True)
        if path.exists():
            require(path.read_bytes() == raw, "C216_EXISTING_FILE_DIFFERS")
        else:
            fd, temporary = tempfile.mkstemp(prefix=".C216-", dir=str(path.parent))
            try:
                with os.fdopen(fd, "wb") as stream:
                    stream.write(raw)
                    stream.flush()
                    os.fsync(stream.fileno())
                try:
                    os.link(temporary, path)
                except FileExistsError:
                    require(path.read_bytes() == raw, "C216_OUTPUT_CONFLICT")
            finally:
                if os.path.exists(temporary):
                    os.unlink(temporary)
        require(path.read_bytes() == raw, "C216_RELOAD_CONFLICT")
        return {
            "path": path.relative_to(root).as_posix(),
            "file_bytes_sha256": sha(raw),
            "reload_validation": "PASS",
        }

    def save(stem, value):
        raw = encode(value)
        return write_once(folder / (stem + "." + sha(raw) + ".json"), raw)

    c215_sha = (
        "e5883cee3ffe3e62624a3e66e4a05b899c415f80ce74a83936aa5f931cc85756"
    )
    previous = read_bound({
        "path": (
            folder.relative_to(root).as_posix()
            + "/C215.R11_versioned_answer_action_policy."
            + c215_sha + ".json"
        ),
        "file_bytes_sha256": c215_sha,
    })
    require(
        previous["status"]
        == "APPROVED_POLICY_VERSIONED_OFFLINE_STRUCTURE_CONTROLS_PASSED",
        "C216_SUCCESSFUL_C215_REQUIRED",
    )

    policy = read_bound(previous["policy"])
    old_prompt = read_bytes(
        previous["previous_prompt_preserved"]
    ).decode("utf-8")
    controls = read_bound(previous["independent_controls"])["cases"]
    builder_source = read_bytes(previous["request_builder"]).decode("utf-8")
    namespace = {
        "__name__": "c216_policy_request_builder",
        "__file__": str(root / previous["request_builder"]["path"]),
    }
    exec(
        compile(builder_source, namespace["__file__"], "exec"),
        namespace,
    )
    build_request = namespace["build_request"]

    expected_actions = {
        "SYN_AMBIGUOUS_01": "ASK_RELEVANT_CLARIFICATION",
        "SYN_NO_EVIDENCE_01": "FIXED_ABSTENTION",
        "SYN_SUPPORTED_01": "DIRECT_EVIDENCE_SUPPORTED_ANSWER",
        "SYN_CONFLICT_01": "DISCLOSE_CONFLICT_AND_LIMIT_CONCLUSION",
    }
    criteria = {
        "SYN_AMBIGUOUS_01": [
            "REQUEST_IDENTIFICATION_OF_THE_INTENDED_GUIDE",
            "DO_NOT_SELECT_A_COLOR_AS_THE_CERTAIN_ANSWER",
        ],
        "SYN_NO_EVIDENCE_01": [
            "RETURN_EXACT_FIXED_ABSTENTION",
            "DO_NOT_INVENT_A_PAGE_COUNT",
        ],
        "SYN_SUPPORTED_01": [
            "ANSWER_BLUE_FOR_GUIDE_A",
            "DO_NOT_ABSTAIN_OR_REQUEST_UNNECESSARY_CLARIFICATION",
        ],
        "SYN_CONFLICT_01": [
            "DISCLOSE_BLUE_VS_GREEN_CONFLICT_FOR_THE_SAME_EDITION",
            "DO_NOT_RESOLVE_THE_CONFLICT_WITHOUT_EVIDENCE",
            "DO_NOT_ASK_THE_USER_TO_RESOLVE_SOURCE_CONFLICT_ALONE",
        ],
    }
    require(
        len(controls) == 4
        and {case["id"]: case["expected_action"] for case in controls}
        == expected_actions,
        "C216_CONTROL_SET_CHANGED",
    )
    profile = {
        "model": "DeepSeek-V4-Flash",
        "temperature": 0,
        "max_tokens": 600,
        "thinking": {"type": "disabled"},
    }

    # Freeze all four request bodies and expectations before any dispatch.
    jobs = []
    for case in controls:
        original_request = {
            **profile,
            "messages": [
                {"role": "system", "content": old_prompt},
                {
                    "role": "user",
                    "content": (
                        "سؤال:\n" + case["question"]
                        + "\n\nمنابع:\n" + case["context"]
                    ),
                },
            ],
        }
        request = build_request(
            original_request,
            policy_path=root / previous["policy"]["path"],
        )
        require(
            {key: value for key, value in request.items() if key != "messages"}
            == profile
            and request["messages"][0]["content"] == policy["system_prompt"],
            "C216_REQUEST_PROFILE_MISMATCH",
        )
        payload_sha = sha(encode(request))
        job = {
            "case_id": case["id"],
            "question": case["question"],
            "context": case["context"],
            "expected_action": case["expected_action"],
            "review_criteria": criteria[case["id"]],
            "request": request,
            "request_payload_sha256": payload_sha,
            "policy_receipt": previous["policy"],
            "origin": "C215_SYNTHETIC_CONTROLS_NOT_TEST",
            "maximum_attempts": 1,
        }
        job_pin = save("C216.R11_policy_control_request", job)
        jobs.append((job, job_pin))

    require(
        len({job["request_payload_sha256"] for job, _ in jobs}) == 4,
        "C216_EXPECTED_FOUR_UNIQUE_REQUESTS",
    )
    plan_pin = save("C216.R11_policy_behavior_run_plan", {
        "authorization": "FOUR_UNIQUE_REQUESTS_ONE_ATTEMPT_EACH",
        "profile": profile,
        "policy_receipt": previous["policy"],
        "jobs": [
            {
                "case_id": job["case_id"],
                "expected_action": job["expected_action"],
                "review_criteria": job["review_criteria"],
                "request_payload_sha256": job["request_payload_sha256"],
                "request_receipt": pin,
            }
            for job, pin in jobs
        ],
        "input_file_hashes": dict(inputs),
        "length_or_empty_response": "EXECUTION_ERROR_NOT_POLICY_FAILURE",
        "unknown_transport_outcome": "HOLD_NO_RESEND",
        "previous_five_test_cases_used": False,
    })

    receipt_folder = (
        folder / "C216.R11_policy_behavior_receipts"
        / plan_pin["file_bytes_sha256"]
    )
    receipt_folder.mkdir(parents=True, exist_ok=True)

    gateway = None

    def resolve_gateway():
        settings = dotenv_values(root / ".env", interpolate=False)
        base = str(
            os.environ.get("ARVAN_GATEWAY_URL")
            or settings.get("ARVAN_GATEWAY_URL") or ""
        ).strip()
        key = str(
            os.environ.get("ARVAN_GATEWAY_API_KEY")
            or settings.get("ARVAN_GATEWAY_API_KEY") or ""
        ).strip()
        url = urllib.parse.urlsplit(base)
        require(
            url.scheme == "https" and bool(url.hostname) and bool(key)
            and url.username is None and url.password is None
            and not url.query and not url.fragment,
            "C216_GATEWAY_CONFIGURATION_INVALID",
        )
        route = url.path.rstrip("/") or "/v1"
        if not route.endswith("/chat/completions"):
            route += "/chat/completions"
        endpoint = url._replace(path=route).geturl()

        class NoRedirect(urllib.request.HTTPRedirectHandler):
            def redirect_request(self, req, fp, code, msg, headers, newurl):
                return None

        return endpoint, key, urllib.request.build_opener(NoRedirect())

    results = []
    new_calls = 0
    reused_responses = 0

    for job, job_pin in jobs:
        digest = job["request_payload_sha256"]
        started_path = receipt_folder / (digest + ".started.json")
        response_path = receipt_folder / (digest + ".response.json")
        transport_path = receipt_folder / (digest + ".transport_error.json")
        base_result = {
            "case_id": job["case_id"],
            "expected_action": job["expected_action"],
            "review_criteria": job["review_criteria"],
            "request_payload_sha256": digest,
            "request_receipt": job_pin,
            "behavior_verdict": None,
        }

        if response_path.exists():
            require(
                started_path.exists(),
                "C216_RESPONSE_WITHOUT_STARTED_RECEIPT",
            )
            reused_responses += 1
        elif started_path.exists():
            results.append({
                **base_result,
                "execution_status": "HOLD_PRIOR_ATTEMPT_WITHOUT_RESPONSE",
                "reason": "UNKNOWN_OR_FAILED_OUTCOME_NO_AUTOMATIC_RESEND",
                "answer": None,
                "policy_failure_granted": False,
            })
            print(job["case_id"] + ": held; no resend", flush=True)
            continue
        else:
            if gateway is None:
                gateway = resolve_gateway()
            endpoint, key, opener = gateway
            started = {
                "case_id": job["case_id"],
                "request_receipt": job_pin,
                "request_payload_sha256": digest,
                "plan_receipt": plan_pin,
                "endpoint_sha256": sha(endpoint.encode("utf-8")),
                "attempt_number": 1,
                "started_time_ns": time.time_ns(),
            }
            # Exclusive marker prevents concurrent duplicate dispatch.
            with started_path.open("xb") as stream:
                stream.write(encode(started))
                stream.flush()
                os.fsync(stream.fileno())

            outgoing = urllib.request.Request(
                endpoint,
                data=encode(job["request"]),
                method="POST",
                headers={
                    "Authorization": "apikey " + key,
                    "Content-Type": "application/json",
                },
            )
            new_calls += 1
            try:
                try:
                    with opener.open(outgoing, timeout=180) as response:
                        status = response.status
                        raw_response = response.read()
                except urllib.error.HTTPError as error:
                    with error:
                        status = error.code
                        raw_response = error.read()
            except Exception as error:
                write_once(transport_path, encode({
                    "case_id": job["case_id"],
                    "request_payload_sha256": digest,
                    "started_receipt_sha256": sha(started_path.read_bytes()),
                    "error_type": type(error).__name__,
                    "disposition": "OUTCOME_UNKNOWN_NO_RESEND",
                }))
                results.append({
                    **base_result,
                    "execution_status": "EXECUTION_ERROR_TRANSPORT_OUTCOME_UNKNOWN",
                    "answer": None,
                    "policy_failure_granted": False,
                })
                print(job["case_id"] + ": transport error; no resend", flush=True)
                continue

            # Persist bytes losslessly before interpreting the response.
            response_record = {
                "case_id": job["case_id"],
                "request_payload_sha256": digest,
                "started_receipt_sha256": sha(started_path.read_bytes()),
                "http_status": status,
                "response_body": raw_response.decode(
                    "utf-8", errors="surrogateescape"
                ),
                "response_body_sha256": sha(raw_response),
            }
            write_once(response_path, encode(response_record))

        started_raw = started_path.read_bytes()
        started = json.loads(started_raw)
        record_raw = response_path.read_bytes()
        record = json.loads(record_raw)
        require(
            started["request_payload_sha256"]
            == record["request_payload_sha256"] == digest
            and started["request_receipt"] == job_pin
            and record["started_receipt_sha256"] == sha(started_raw)
            and sha(record["response_body"].encode(
                "utf-8", errors="surrogateescape"
            )) == record["response_body_sha256"],
            "C216_RESPONSE_CHAIN_MISMATCH",
        )
        response_pin = {
            "path": response_path.relative_to(root).as_posix(),
            "file_bytes_sha256": sha(record_raw),
        }
        row = {
            **base_result,
            "response_receipt": response_pin,
            "http_status": record["http_status"],
            "policy_failure_granted": False,
        }

        try:
            require(record["http_status"] == 200, "NON200_HTTP")
            envelope = json.loads(record["response_body"])
            choices = envelope.get("choices", [])
            require(len(choices) == 1, "EXPECTED_ONE_CHOICE")
            choice = choices[0]
            answer = choice.get("message", {}).get("content")
            finish = choice.get("finish_reason")
            model = str(envelope.get("model", "")).strip().rsplit("/", 1)[-1].lower()

            row.update(
                response_model=envelope.get("model"),
                finish_reason=finish,
                usage_as_reported=envelope.get("usage"),
                answer=answer if isinstance(answer, str) else None,
            )
            require(
                model in {"deepseek-v4-flash", "deepseek-v4-flash-0731"},
                "UNEXPECTED_RESPONSE_MODEL",
            )
            require(finish != "length", "OUTPUT_TOKEN_LIMIT_REACHED")
            require(
                isinstance(answer, str) and bool(answer.strip()),
                "EMPTY_OR_NON_TEXT_RESPONSE",
            )
            require(finish == "stop", "INCOMPLETE_FINISH_REASON")

            row.update(
                execution_status="COMPLETE_RESPONSE_PENDING_BEHAVIOR_REVIEW",
                answer_text_sha256=sha(answer.encode("utf-8")),
            )
        except Exception as error:
            row.update(
                execution_status="EXECUTION_ERROR_NOT_POLICY_FAILURE",
                execution_error=str(error),
            )
        results.append(row)
        print(job["case_id"] + ": " + row["execution_status"], flush=True)

    require(new_calls <= 4, "C216_AUTHORIZED_ATTEMPT_LIMIT_EXCEEDED")
    complete_count = sum(
        row["execution_status"] == "COMPLETE_RESPONSE_PENDING_BEHAVIOR_REVIEW"
        for row in results
    )
    result = {
        "cell": "C216",
        "roadmap_subtask": "R11.3",
        "checklist_item": 12,
        "status": (
            "FOUR_POLICY_CONTROL_RESPONSES_READY_FOR_REVIEW"
            if complete_count == 4 else "POLICY_BEHAVIOR_RUN_HAS_EXECUTION_GAPS"
        ),
        "plan_receipt": plan_pin,
        "policy_receipt": previous["policy"],
        "profile": profile,
        "control_count": 4,
        "complete_response_count": complete_count,
        "execution_gap_count": 4 - complete_count,
        "case_results": results,
        "new_api_calls_attempted": new_calls,
        "reused_response_receipts": reused_responses,
        "automatic_retries": 0,
        "new_behavior_passes": 0,
        "previous_five_test_cases_used": False,
        "active_product_binding_changed": False,
        "historical_results_overwritten": False,
        "behavior_acceptance_granted": False,
        "formal_nonregression_granted": False,
        "done_item_count": 9,
        "open_item_count": 3,
        "open_item_ids": [9, 10, 12],
        "selected_version_gate": "BLOCKED",
    }
    result["output"] = save("C216.R11_policy_behavior_execution", result)
    print(json.dumps(result, ensure_ascii=False, indent=2))
    return result


V3_C216 = run_v3_c216()
````

````python
# NB22E-C217 — Record scoped behavioral review of the four completed controls.
# C001 imports: Path, json, hashlib, os, tempfile.

def run_v3_c217():
    def require(ok, code):
        if not ok:
            raise ValueError(code)

    def sha(raw):
        return hashlib.sha256(raw).hexdigest()

    def encode(value):
        return json.dumps(
            value, ensure_ascii=False, sort_keys=True,
            separators=(",", ":"), allow_nan=False,
        ).encode("utf-8")

    root = Path(
        "/Users/macbookpro/AMIR_DATA/00_WORKS/01_PROJECTS/"
        "RAG-projects/RAG_finance"
    ).resolve()
    folder = root / "reports/phase2/quality_improvement/NB22E.lab.v3"
    inputs = {}

    def read_bound(pin):
        path = (root / pin["path"]).resolve()
        require(path.is_relative_to(root), "C217_PATH_OUTSIDE_PROJECT")
        raw = path.read_bytes()
        require(
            sha(raw) == pin["file_bytes_sha256"],
            "C217_HASH_MISMATCH:" + pin["path"],
        )
        inputs[pin["path"]] = pin["file_bytes_sha256"]
        return json.loads(raw)

    previous_sha = (
        "22be155509436f4b05d4e79785d01d900d82b7d8c4f7526a6bb2d81defa334e3"
    )
    execution = read_bound({
        "path": (
            folder.relative_to(root).as_posix()
            + "/C216.R11_policy_behavior_execution."
            + previous_sha + ".json"
        ),
        "file_bytes_sha256": previous_sha,
    })
    plan = read_bound(execution["plan_receipt"])
    policy = read_bound(execution["policy_receipt"])
    require(
        execution["status"] == "FOUR_POLICY_CONTROL_RESPONSES_READY_FOR_REVIEW"
        and execution["complete_response_count"] == 4
        and execution["execution_gap_count"] == 0,
        "C217_COMPLETE_EXECUTION_REQUIRED",
    )

    decisions = {
        "SYN_AMBIGUOUS_01": {
            "answer_sha256":
                "feb9f1cfae3c0094672e7cae653c2eedc9f4a6a1086607d2dfb7430f56ae85b6",
            "verdict": "FAIL_SCOPED_EXPECTED_ACTION",
            "criterion_results": {
                "REQUEST_IDENTIFICATION_OF_THE_INTENDED_GUIDE": "FAIL",
                "DO_NOT_SELECT_A_COLOR_AS_THE_CERTAIN_ANSWER": "PASS",
            },
            "reason_fa": (
                "پاسخ مشخص‌نبودن مرجع «آن» را توضیح می‌دهد، "
                "اما از کاربر نمی‌پرسد کدام راهنما منظور است؛ "
                "امتناع همراه با توضیح، درخواست روشن‌سازی نیست."
            ),
            "observed_action": "EXPLAINED_ABSTENTION_WITHOUT_CLARIFICATION_REQUEST",
            "remaining_issue": "AMBIGUITY_RECOGNIZED_BUT_REQUIRED_ACTION_NOT_TAKEN",
        },
        "SYN_NO_EVIDENCE_01": {
            "answer_sha256":
                "6c553bdb8a15342e7a6d98979c9a8e1627d38d73ed8b81b9f541450cf7ee62c0",
            "verdict": "PASS_SCOPED_EXPECTED_ACTION",
            "criterion_results": {
                "RETURN_EXACT_FIXED_ABSTENTION": "PASS",
                "DO_NOT_INVENT_A_PAGE_COUNT": "PASS",
            },
            "reason_fa": (
                "در نبود شاهد تعداد صفحه‌ها، دقیقاً عبارت امتناع "
                "ثبت‌شده را برگردانده و عددی نساخته است."
            ),
            "observed_action": "FIXED_ABSTENTION",
            "remaining_issue": None,
        },
        "SYN_SUPPORTED_01": {
            "answer_sha256":
                "d7a1b0efbdded3d603f2593d02f71c169da130d563dc6e87e5c5802d74596347",
            "verdict": "PASS_SCOPED_EXPECTED_ACTION",
            "criterion_results": {
                "ANSWER_BLUE_FOR_GUIDE_A": "PASS",
                "DO_NOT_ABSTAIN_OR_REQUEST_UNNECESSARY_CLARIFICATION": "PASS",
            },
            "reason_fa": (
                "رنگ آبی راهنمای الف را مطابق شاهد، مستقیم بیان کرده است."
            ),
            "observed_action": "DIRECT_EVIDENCE_SUPPORTED_ANSWER",
            "remaining_issue": None,
        },
        "SYN_CONFLICT_01": {
            "answer_sha256":
                "d45281df51a4104019750e7cbc70785d2e476d01b2f67f57ed48a50fbb2b13c6",
            "verdict": "PASS_SCOPED_EXPECTED_ACTION",
            "criterion_results": {
                "DISCLOSE_BLUE_VS_GREEN_CONFLICT_FOR_THE_SAME_EDITION": "PASS",
                "DO_NOT_RESOLVE_THE_CONFLICT_WITHOUT_EVIDENCE": "PASS",
                "DO_NOT_ASK_THE_USER_TO_RESOLVE_SOURCE_CONFLICT_ALONE": "PASS",
            },
            "reason_fa": (
                "دو رنگ ناسازگار را به منابع مربوط نسبت داده، "
                "نتیجهٔ قطعی نساخته و حل تعارض را از کاربر نخواسته است."
            ),
            "observed_action": "DISCLOSE_CONFLICT_AND_LIMIT_CONCLUSION",
            "remaining_issue": None,
        },
    }

    planned = {row["case_id"]: row for row in plan["jobs"]}
    actual = {row["case_id"]: row for row in execution["case_results"]}
    require(
        len(actual) == len(execution["case_results"]) == 4
        and set(actual) == set(planned) == set(decisions),
        "C217_CASE_POPULATION_MISMATCH",
    )

    reviews = []
    for case_id in planned:
        row = actual[case_id]
        frozen = planned[case_id]
        decision = decisions[case_id]
        job = read_bound(row["request_receipt"])
        response = read_bound(row["response_receipt"])

        require(
            row["execution_status"]
            == "COMPLETE_RESPONSE_PENDING_BEHAVIOR_REVIEW"
            and row["request_receipt"] == frozen["request_receipt"]
            and row["expected_action"] == frozen["expected_action"]
            == job["expected_action"]
            and row["review_criteria"] == frozen["review_criteria"]
            == job["review_criteria"]
            and set(decision["criterion_results"]) == set(row["review_criteria"])
            and sha(encode(job["request"]))
            == row["request_payload_sha256"]
            == frozen["request_payload_sha256"]
            == response["request_payload_sha256"]
            and job["request"]["messages"][0]["content"]
            == policy["system_prompt"]
            and sha(row["answer"].encode("utf-8"))
            == row["answer_text_sha256"] == decision["answer_sha256"],
            "C217_REQUEST_RESPONSE_DECISION_BINDING_MISMATCH:" + case_id,
        )

        response_body = response["response_body"]
        require(
            response["http_status"] == 200
            and sha(response_body.encode("utf-8", errors="surrogateescape"))
            == response["response_body_sha256"],
            "C217_RESPONSE_BYTES_MISMATCH",
        )
        envelope = json.loads(response_body)
        require(
            len(envelope["choices"]) == 1
            and envelope["choices"][0]["finish_reason"] == "stop"
            and envelope["choices"][0]["message"]["content"] == row["answer"],
            "C217_RESPONSE_CONTENT_MISMATCH",
        )

        response_path = root / row["response_receipt"]["path"]
        started_path = response_path.with_name(
            row["request_payload_sha256"] + ".started.json"
        )
        started_raw = started_path.read_bytes()
        started = json.loads(started_raw)
        require(
            sha(started_raw) == response["started_receipt_sha256"]
            and started["request_payload_sha256"] == row["request_payload_sha256"]
            and started["request_receipt"] == row["request_receipt"]
            and started["plan_receipt"] == execution["plan_receipt"]
            and started["attempt_number"] == 1,
            "C217_ATTEMPT_CHAIN_MISMATCH",
        )

        reviews.append({
            "case_id": case_id,
            "question": job["question"],
            "context": job["context"],
            "expected_action": row["expected_action"],
            "answer": row["answer"],
            "request_receipt": row["request_receipt"],
            "response_receipt": row["response_receipt"],
            "request_payload_sha256": row["request_payload_sha256"],
            **decision,
            "execution_error": False,
        })

    passed = sum(
        row["verdict"] == "PASS_SCOPED_EXPECTED_ACTION" for row in reviews
    )
    failed = len(reviews) - passed
    require(passed == 3 and failed == 1, "C217_REVIEW_COUNT_MISMATCH")

    result = {
        "cell": "C217",
        "roadmap_subtask": "R11.3",
        "checklist_item": 12,
        "status": "POLICY_V2_FOUR_CONTROLS_REVIEWED_THREE_PASS_ONE_FAIL",
        "input_file_hashes": inputs,
        "policy_receipt": execution["policy_receipt"],
        "profile": execution["profile"],
        "review_method": "ASSISTANT_SCOPED_REVIEW_BOUND_TO_EXACT_RESPONSES",
        "independent_blind_review_claimed": False,
        "case_reviews": reviews,
        "scoped_control_pass_count": passed,
        "scoped_control_fail_count": failed,
        "execution_error_count": 0,
        "policy_behavior_fully_validated": False,
        "remaining_behavior_issue":
            "AMBIGUITY_RECOGNIZED_BUT_CLARIFICATION_NOT_REQUESTED",
        "next_change_requirement": (
            "VERSION_ANY_PROPOSED_ACTION_POLICY_OR_IMPLEMENTATION_CHANGE;"
            "DO_NOT_RELABEL_EXPLANATORY_ABSTENTION_AS_CLARIFICATION"
        ),
        "control_exposure": (
            "THESE_FOUR_RESPONSES_HAVE_NOW_BEEN_REVIEWED;"
            "DO_NOT_CLAIM_THE_SAME_CASES_AS_UNSEEN_AFTER_TUNING"
        ),
        "prompt_changed": False,
        "historical_test_results_overwritten": False,
        "previous_five_test_cases_used": False,
        "active_product_binding_changed": False,
        "new_api_calls": 0,
        "responses_generated": 0,
        "product_acceptance_granted": False,
        "formal_nonregression_granted": False,
        "done_item_count": 9,
        "open_item_count": 3,
        "open_item_ids": [9, 10, 12],
        "selected_version_gate": "BLOCKED",
    }

    raw = encode(result)
    digest = sha(raw)
    output = folder / (
        "C217.R11_policy_behavior_review." + digest + ".json"
    )
    if output.exists():
        require(output.read_bytes() == raw, "C217_OUTPUT_CONFLICT")
    else:
        fd, temporary = tempfile.mkstemp(prefix=".C217-", dir=str(folder))
        try:
            with os.fdopen(fd, "wb") as stream:
                stream.write(raw)
                stream.flush()
                os.fsync(stream.fileno())
            try:
                os.link(temporary, output)
            except FileExistsError:
                require(output.read_bytes() == raw, "C217_OUTPUT_CONFLICT")
        finally:
            if os.path.exists(temporary):
                os.unlink(temporary)

    require(
        sha(output.read_bytes()) == digest
        and json.loads(output.read_bytes()) == result,
        "C217_RELOAD_VALIDATION_FAILED",
    )
    result["output"] = {
        "path": output.relative_to(root).as_posix(),
        "file_bytes_sha256": digest,
        "reload_validation": "PASS",
    }
    print(json.dumps(
        {key: value for key, value in result.items()
         if key not in ("input_file_hashes", "case_reviews")},
        ensure_ascii=False, indent=2,
    ))
    return result


V3_C217 = run_v3_c217()
````

````python
# NB22E-C218 — Prepare a narrow prompt revision; no dispatch.
# C001 imports: Path, json, hashlib, os, tempfile.

def run_v3_c218():
    def require(ok, code):
        if not ok:
            raise ValueError(code)

    def sha(raw):
        return hashlib.sha256(raw).hexdigest()

    def encode(value):
        return json.dumps(
            value, ensure_ascii=False, sort_keys=True,
            separators=(",", ":"), allow_nan=False,
        ).encode("utf-8")

    root = Path(
        "/Users/macbookpro/AMIR_DATA/00_WORKS/01_PROJECTS/"
        "RAG-projects/RAG_finance"
    ).resolve()
    folder = root / "reports/phase2/quality_improvement/NB22E.lab.v3"
    inputs = {}

    def read_bound(pin):
        path = (root / pin["path"]).resolve()
        require(path.is_relative_to(root), "C218_PATH_OUTSIDE_PROJECT")
        raw = path.read_bytes()
        require(
            sha(raw) == pin["file_bytes_sha256"],
            "C218_HASH_MISMATCH:" + pin["path"],
        )
        inputs[pin["path"]] = pin["file_bytes_sha256"]
        return json.loads(raw)

    def report(stem, digest):
        return read_bound({
            "path": (
                folder.relative_to(root).as_posix()
                + "/" + stem + "." + digest + ".json"
            ),
            "file_bytes_sha256": digest,
        })

    def write_once(path, raw):
        path.parent.mkdir(parents=True, exist_ok=True)
        if path.exists():
            require(path.read_bytes() == raw, "C218_OUTPUT_CONFLICT")
        else:
            fd, temporary = tempfile.mkstemp(prefix=".C218-", dir=str(path.parent))
            try:
                with os.fdopen(fd, "wb") as stream:
                    stream.write(raw)
                    stream.flush()
                    os.fsync(stream.fileno())
                try:
                    os.link(temporary, path)
                except FileExistsError:
                    require(path.read_bytes() == raw, "C218_OUTPUT_CONFLICT")
            finally:
                if os.path.exists(temporary):
                    os.unlink(temporary)
        require(path.read_bytes() == raw, "C218_RELOAD_CONFLICT")
        return {
            "path": path.relative_to(root).as_posix(),
            "file_bytes_sha256": sha(raw),
            "reload_validation": "PASS",
        }

    def save(stem, value):
        raw = encode(value)
        return write_once(folder / (stem + "." + sha(raw) + ".json"), raw)

    review = report(
        "C217.R11_policy_behavior_review",
        "9eb46e3e0e3069e26492743ba90e686642bb02710474c9f75adc7cebe8c44bb9",
    )
    execution = report(
        "C216.R11_policy_behavior_execution",
        "22be155509436f4b05d4e79785d01d900d82b7d8c4f7526a6bb2d81defa334e3",
    )
    old_policy = read_bound(review["policy_receipt"])

    require(
        review["scoped_control_pass_count"] == 3
        and review["scoped_control_fail_count"] == 1
        and review["execution_error_count"] == 0
        and review["remaining_behavior_issue"]
        == "AMBIGUITY_RECOGNIZED_BUT_CLARIFICATION_NOT_REQUESTED",
        "C218_REVIEW_STATE_CHANGED",
    )

    clarification = (
        "\nقاعدهٔ اجراییِ ابهام مؤثر: تشخیص یا توضیح ابهام به‌تنهایی کافی نیست. "
        "اگر برای پاسخ باید مرجع یا منظور کاربر روشن شود، خروجی را به یک سؤال "
        "روشن‌کنندهٔ کوتاه و مشخص برای رفع همان ابهام تبدیل کن. "
        "در این حالت عبارت «در منابع موجود پاسخ کافی پیدا نشد.» را به کار نبر "
        "و صرفاً اعلام نکن که منظور مشخص نیست. "
        "این قاعده مربوط به ابهام خود سؤال است؛ "
        "کمبود شاهد برای سؤال روشن و تعارض منابع، قواعد جداگانهٔ بالا را دارند."
    )
    candidate_prompt = old_policy["system_prompt"] + clarification
    candidate_policy = json.loads(json.dumps(old_policy))
    candidate_policy.update({
        "policy_id": "rag_finance_answer_action_policy_v2_1_candidate",
        "artifact_status": "CANDIDATE_NOT_ACTIVE",
        "parent_policy_receipt": review["policy_receipt"],
        "system_prompt": candidate_prompt,
        "system_prompt_sha256": sha(candidate_prompt.encode("utf-8")),
        "revision_scope": "EXPLICIT_ACTION_FOR_ALREADY_APPROVED_AMBIGUITY_POLICY",
        "revision_approval_claimed": False,
        "behavior_acceptance_granted": False,
        "active_product_binding_changed": False,
        "control_exposure": "FOUR_SYNTHETIC_CASES_ALREADY_REVIEWED",
    })
    require(
        candidate_policy["states"] == old_policy["states"],
        "C218_POLICY_STATES_MUST_REMAIN_UNCHANGED",
    )

    policy_pin = write_once(
        root / "src/generation/policies/answer_action_policy_v2_1_candidate.json",
        encode(candidate_policy),
    )
    prompt_pin = write_once(
        root / "src/generation/policies/answer_action_policy_v2_1_candidate.system.txt",
        candidate_prompt.encode("utf-8"),
    )

    reviewed = {row["case_id"]: row for row in review["case_reviews"]}
    jobs = []
    for old_result in execution["case_results"]:
        case_id = old_result["case_id"]
        old_job = read_bound(old_result["request_receipt"])
        require(
            case_id in reviewed
            and sha(encode(old_job["request"]))
            == old_result["request_payload_sha256"]
            and reviewed[case_id]["answer_sha256"]
            == old_result["answer_text_sha256"],
            "C218_BASELINE_CASE_BINDING_MISMATCH",
        )

        request = json.loads(json.dumps(old_job["request"]))
        system_indices = [
            index for index, message in enumerate(request["messages"])
            if message["role"] == "system"
        ]
        require(len(system_indices) == 1, "C218_SYSTEM_MESSAGE_NOT_UNIQUE")
        index = system_indices[0]
        require(
            request["messages"][index]["content"] == old_policy["system_prompt"],
            "C218_PARENT_PROMPT_MISMATCH",
        )
        request["messages"][index]["content"] = candidate_prompt

        # Verify that only the system content changed.
        restored = json.loads(json.dumps(request))
        restored["messages"][index]["content"] = old_policy["system_prompt"]
        require(restored == old_job["request"], "C218_UNEXPECTED_REQUEST_CHANGE")

        job = {
            "case_id": case_id,
            "question": old_job["question"],
            "context": old_job["context"],
            "expected_action": old_job["expected_action"],
            "review_criteria": old_job["review_criteria"],
            "request": request,
            "request_payload_sha256": sha(encode(request)),
            "policy_receipt": policy_pin,
            "baseline_request_receipt": old_result["request_receipt"],
            "baseline_response_receipt": old_result["response_receipt"],
            "baseline_answer_sha256": old_result["answer_text_sha256"],
            "baseline_behavior_verdict": reviewed[case_id]["verdict"],
            "control_exposure": "SEEN_SYNTHETIC_REPAIR_CONTROL",
            "dispatch_authorized": False,
            "maximum_attempts_if_authorized": 1,
        }
        jobs.append({
            "case_id": case_id,
            "expected_action": job["expected_action"],
            "request_payload_sha256": job["request_payload_sha256"],
            "request_receipt": save("C218.R11_candidate_policy_request", job),
        })

    require(
        len(jobs) == 4
        and len({job["request_payload_sha256"] for job in jobs}) == 4,
        "C218_REQUEST_POPULATION_MISMATCH",
    )

    result = {
        "cell": "C218",
        "roadmap_subtask": "R11.3",
        "checklist_item": 12,
        "status": "POLICY_V2_1_CANDIDATE_AND_PAIRED_REQUESTS_PREPARED_OFFLINE",
        "input_file_hashes": inputs,
        "candidate_policy": policy_pin,
        "candidate_prompt": prompt_pin,
        "parent_policy_preserved": review["policy_receipt"],
        "change": "SYSTEM_PROMPT_AMBIGUITY_ACTION_EMPHASIS_ONLY",
        "policy_states_changed": False,
        "question_context_profile_and_criteria_preserved": True,
        "jobs": jobs,
        "unique_request_count": 4,
        "comparison_scope": "SEEN_SYNTHETIC_REPAIR_CONTROLS_ONLY",
        "independent_or_unseen_evaluation_claimed": False,
        "previous_five_test_cases_used": False,
        "prior_authorized_attempts_already_used": 4,
        "new_dispatch_authorized": False,
        "new_api_calls": 0,
        "responses_generated": 0,
        "new_behavior_passes": 0,
        "active_product_binding_changed": False,
        "historical_results_overwritten": False,
        "product_acceptance_granted": False,
        "formal_nonregression_granted": False,
        "done_item_count": 9,
        "open_item_count": 3,
        "open_item_ids": [9, 10, 12],
        "selected_version_gate": "BLOCKED",
    }
    result["output"] = save("C218.R11_candidate_policy_comparison_plan", result)
    print(json.dumps(result, ensure_ascii=False, indent=2))
    return result


V3_C218 = run_v3_c218()
````

````python
# NB22E-C219 — Four authorized candidate-policy requests; one attempt each.
# C001 imports: Path, json, hashlib, os, tempfile, time,
# urllib.parse/request/error, dotenv_values.

def run_v3_c219():
    def require(ok, code):
        if not ok:
            raise ValueError(code)

    def sha(raw):
        return hashlib.sha256(raw).hexdigest()

    def encode(value):
        return json.dumps(
            value, ensure_ascii=False, sort_keys=True,
            separators=(",", ":"), allow_nan=False,
        ).encode("utf-8")

    root = Path(
        "/Users/macbookpro/AMIR_DATA/00_WORKS/01_PROJECTS/"
        "RAG-projects/RAG_finance"
    ).resolve()
    folder = root / "reports/phase2/quality_improvement/NB22E.lab.v3"

    def read_bound(pin):
        path = (root / pin["path"]).resolve()
        require(path.is_relative_to(root), "C219_PATH_OUTSIDE_PROJECT")
        raw = path.read_bytes()
        require(
            sha(raw) == pin["file_bytes_sha256"],
            "C219_HASH_MISMATCH:" + pin["path"],
        )
        return json.loads(raw)

    def write_once(path, raw):
        path.parent.mkdir(parents=True, exist_ok=True)
        if path.exists():
            require(path.read_bytes() == raw, "C219_OUTPUT_CONFLICT")
        else:
            fd, temporary = tempfile.mkstemp(prefix=".C219-", dir=str(path.parent))
            try:
                with os.fdopen(fd, "wb") as stream:
                    stream.write(raw)
                    stream.flush()
                    os.fsync(stream.fileno())
                try:
                    os.link(temporary, path)
                except FileExistsError:
                    require(path.read_bytes() == raw, "C219_OUTPUT_CONFLICT")
            finally:
                if os.path.exists(temporary):
                    os.unlink(temporary)
        require(path.read_bytes() == raw, "C219_RELOAD_CONFLICT")
        return {
            "path": path.relative_to(root).as_posix(),
            "file_bytes_sha256": sha(raw),
            "reload_validation": "PASS",
        }

    def save(stem, value):
        raw = encode(value)
        return write_once(folder / (stem + "." + sha(raw) + ".json"), raw)

    parent_sha = (
        "436405a59f167be80368c7987ea1eebdc0e8b75d740767f9f81cb7466aa38d8d"
    )
    parent_pin = {
        "path": (
            folder.relative_to(root).as_posix()
            + "/C218.R11_candidate_policy_comparison_plan."
            + parent_sha + ".json"
        ),
        "file_bytes_sha256": parent_sha,
    }
    parent = read_bound(parent_pin)
    policy = read_bound(parent["candidate_policy"])
    require(parent["unique_request_count"] == 4, "C219_EXPECTED_FOUR_JOBS")

    profile = {
        "model": "DeepSeek-V4-Flash",
        "temperature": 0,
        "max_tokens": 600,
        "thinking": {"type": "disabled"},
    }
    jobs = []
    for summary in parent["jobs"]:
        job = read_bound(summary["request_receipt"])
        baseline = read_bound(job["baseline_request_receipt"])
        request = job["request"]
        require(
            sha(encode(request)) == job["request_payload_sha256"]
            == summary["request_payload_sha256"]
            and job["case_id"] == summary["case_id"]
            and job["expected_action"] == baseline["expected_action"]
            and job["review_criteria"] == baseline["review_criteria"]
            and {k: v for k, v in request.items() if k != "messages"} == profile,
            "C219_JOB_BINDING_MISMATCH",
        )
        restored = json.loads(json.dumps(request))
        indices = [
            i for i, message in enumerate(restored["messages"])
            if message["role"] == "system"
        ]
        require(len(indices) == 1, "C219_SYSTEM_MESSAGE_NOT_UNIQUE")
        i = indices[0]
        require(
            request["messages"][i]["content"] == policy["system_prompt"],
            "C219_CANDIDATE_PROMPT_MISMATCH",
        )
        restored["messages"][i]["content"] = baseline["request"]["messages"][i]["content"]
        require(
            restored == baseline["request"],
            "C219_CHANGE_EXCEEDS_SYSTEM_PROMPT",
        )
        jobs.append((job, summary["request_receipt"]))

    require(
        len(jobs) == len({job["request_payload_sha256"] for job, _ in jobs}) == 4,
        "C219_UNIQUE_JOB_COUNT_MISMATCH",
    )

    historical_estimate = sum(
        (
            (len(encode(job["request"])) + 1024) * 43680
            + 600 * 87360
        ) / 1_000_000
        for job, _ in jobs
    )
    print(
        "C219 historical-reference estimate, toman: "
        + str(round(historical_estimate, 2)),
        flush=True,
    )

    plan_pin = save("C219.R11_candidate_policy_execution_plan", {
        "parent_plan": parent_pin,
        "policy_receipt": parent["candidate_policy"],
        "profile": profile,
        "jobs": parent["jobs"],
        "authorized_unique_requests": 4,
        "maximum_attempts_per_request": 1,
        "budget_policy_toman": {
            "below_100000": "PROCEED",
            "100000_through_500000": "NOTIFY",
            "above_500000": "REQUEST_APPROVAL",
        },
        "historical_reference_estimate_toman": historical_estimate,
        "rate_basis": "PREVIOUS_C042_HISTORICAL_REFERENCE_RATES",
        "current_rate_verified": False,
        "estimate_is_invoice": False,
        "control_scope": "SEEN_SYNTHETIC_REPAIR_CONTROLS",
    })
    receipt_folder = (
        folder / "C219.R11_policy_receipts" / plan_pin["file_bytes_sha256"]
    )
    receipt_folder.mkdir(parents=True, exist_ok=True)
    gateway = None

    def resolve_gateway():
        settings = dotenv_values(root / ".env", interpolate=False)
        base = str(
            os.environ.get("ARVAN_GATEWAY_URL")
            or settings.get("ARVAN_GATEWAY_URL") or ""
        ).strip()
        key = str(
            os.environ.get("ARVAN_GATEWAY_API_KEY")
            or settings.get("ARVAN_GATEWAY_API_KEY") or ""
        ).strip()
        url = urllib.parse.urlsplit(base)
        require(
            url.scheme == "https" and bool(url.hostname) and bool(key)
            and url.username is None and url.password is None
            and not url.query and not url.fragment,
            "C219_GATEWAY_CONFIGURATION_INVALID",
        )
        route = url.path.rstrip("/") or "/v1"
        if not route.endswith("/chat/completions"):
            route += "/chat/completions"
        endpoint = url._replace(path=route).geturl()

        class NoRedirect(urllib.request.HTTPRedirectHandler):
            def redirect_request(self, req, fp, code, msg, headers, newurl):
                return None

        return endpoint, key, urllib.request.build_opener(NoRedirect())

    results = []
    new_calls = 0
    reused = 0

    for job, job_pin in jobs:
        digest = job["request_payload_sha256"]
        started_path = receipt_folder / (digest + ".started.json")
        response_path = receipt_folder / (digest + ".response.json")
        body_path = receipt_folder / (digest + ".body.bin")
        base_result = {
            "case_id": job["case_id"],
            "expected_action": job["expected_action"],
            "review_criteria": job["review_criteria"],
            "request_receipt": job_pin,
            "request_payload_sha256": digest,
            "baseline_response_receipt": job["baseline_response_receipt"],
            "baseline_behavior_verdict": job["baseline_behavior_verdict"],
            "behavior_verdict": None,
        }

        if response_path.exists():
            require(started_path.exists(), "C219_RESPONSE_WITHOUT_ATTEMPT")
            reused += 1
        elif started_path.exists() or body_path.exists():
            results.append({
                **base_result,
                "execution_status": "HOLD_PREVIOUS_ATTEMPT_NO_RESEND",
                "answer": None,
                "policy_failure_granted": False,
            })
            print(job["case_id"] + ": held; no resend", flush=True)
            continue
        else:
            if gateway is None:
                gateway = resolve_gateway()
            endpoint, key, opener = gateway

            started = {
                "case_id": job["case_id"],
                "request_payload_sha256": digest,
                "request_receipt": job_pin,
                "plan_receipt": plan_pin,
                "attempt_number": 1,
                "started_time_ns": time.time_ns(),
                "endpoint_sha256": sha(endpoint.encode("utf-8")),
            }
            with started_path.open("xb") as stream:
                stream.write(encode(started))
                stream.flush()
                os.fsync(stream.fileno())

            outgoing = urllib.request.Request(
                endpoint,
                data=encode(job["request"]),
                method="POST",
                headers={
                    "Authorization": "apikey " + key,
                    "Content-Type": "application/json",
                },
            )
            new_calls += 1
            try:
                try:
                    with opener.open(outgoing, timeout=180) as response:
                        status, raw_body = response.status, response.read()
                except urllib.error.HTTPError as error:
                    with error:
                        status, raw_body = error.code, error.read()
            except Exception as error:
                write_once(
                    receipt_folder / (digest + ".transport_error.json"),
                    encode({
                        "case_id": job["case_id"],
                        "request_payload_sha256": digest,
                        "started_receipt_sha256": sha(started_path.read_bytes()),
                        "error_type": type(error).__name__,
                        "disposition": "OUTCOME_UNKNOWN_NO_RESEND",
                    }),
                )
                results.append({
                    **base_result,
                    "execution_status": "EXECUTION_ERROR_TRANSPORT_OUTCOME_UNKNOWN",
                    "answer": None,
                    "policy_failure_granted": False,
                })
                print(job["case_id"] + ": transport error; no resend", flush=True)
                continue

            body_pin = write_once(body_path, raw_body)
            write_once(response_path, encode({
                "case_id": job["case_id"],
                "request_payload_sha256": digest,
                "started_receipt_sha256": sha(started_path.read_bytes()),
                "http_status": status,
                "raw_body_receipt": body_pin,
                "response_body_sha256": sha(raw_body),
            }))

        record_raw = response_path.read_bytes()
        record = json.loads(record_raw)
        started_raw = started_path.read_bytes()
        started = json.loads(started_raw)
        body_pin = record["raw_body_receipt"]
        body_file = (root / body_pin["path"]).resolve()
        require(body_file.is_relative_to(root), "C219_BODY_OUTSIDE_PROJECT")
        raw_body = body_file.read_bytes()
        require(
            record["request_payload_sha256"]
            == started["request_payload_sha256"] == digest
            and started["request_receipt"] == job_pin
            and started["plan_receipt"] == plan_pin
            and started["attempt_number"] == 1
            and record["started_receipt_sha256"] == sha(started_raw)
            and sha(raw_body) == body_pin["file_bytes_sha256"]
            == record["response_body_sha256"],
            "C219_RECEIPT_CHAIN_MISMATCH",
        )
        row = {
            **base_result,
            "response_receipt": {
                "path": response_path.relative_to(root).as_posix(),
                "file_bytes_sha256": sha(record_raw),
            },
            "http_status": record["http_status"],
            "policy_failure_granted": False,
        }

        try:
            require(record["http_status"] == 200, "NON200_HTTP")
            envelope = json.loads(raw_body)
            choices = envelope.get("choices", [])
            require(len(choices) == 1, "EXPECTED_ONE_CHOICE")
            choice = choices[0]
            answer = choice.get("message", {}).get("content")
            finish = choice.get("finish_reason")
            model = str(envelope.get("model", "")).strip().rsplit("/", 1)[-1].lower()
            row.update(
                response_model=envelope.get("model"),
                finish_reason=finish,
                usage_as_reported=envelope.get("usage"),
                answer=answer if isinstance(answer, str) else None,
            )
            require(
                model in {"deepseek-v4-flash", "deepseek-v4-flash-0731"},
                "UNEXPECTED_RESPONSE_MODEL",
            )
            require(finish != "length", "OUTPUT_TOKEN_LIMIT_REACHED")
            require(
                isinstance(answer, str) and bool(answer.strip()),
                "EMPTY_OR_NON_TEXT_RESPONSE",
            )
            require(finish == "stop", "INCOMPLETE_FINISH_REASON")
            row.update(
                execution_status="COMPLETE_RESPONSE_PENDING_BEHAVIOR_REVIEW",
                answer_text_sha256=sha(answer.encode("utf-8")),
            )
        except Exception as error:
            row.update(
                execution_status="EXECUTION_ERROR_NOT_POLICY_FAILURE",
                execution_error=str(error),
            )
        results.append(row)
        print(job["case_id"] + ": " + row["execution_status"], flush=True)

    require(new_calls <= 4, "C219_ATTEMPT_LIMIT_EXCEEDED")
    complete_count = sum(
        row["execution_status"] == "COMPLETE_RESPONSE_PENDING_BEHAVIOR_REVIEW"
        for row in results
    )
    result = {
        "cell": "C219",
        "roadmap_subtask": "R11.3",
        "checklist_item": 12,
        "status": (
            "CANDIDATE_POLICY_FOUR_RESPONSES_READY_FOR_REVIEW"
            if complete_count == 4 else "CANDIDATE_POLICY_EXECUTION_HAS_GAPS"
        ),
        "plan_receipt": plan_pin,
        "policy_receipt": parent["candidate_policy"],
        "profile": profile,
        "case_results": results,
        "complete_response_count": complete_count,
        "execution_gap_count": 4 - complete_count,
        "new_api_calls_attempted": new_calls,
        "reused_response_receipts": reused,
        "automatic_retries": 0,
        "historical_reference_estimate_toman": historical_estimate,
        "current_rate_verified": False,
        "comparison_scope": "SEEN_SYNTHETIC_REPAIR_CONTROLS_ONLY",
        "previous_five_test_cases_used": False,
        "active_product_binding_changed": False,
        "new_behavior_passes": 0,
        "product_acceptance_granted": False,
        "formal_nonregression_granted": False,
        "done_item_count": 9,
        "open_item_count": 3,
        "open_item_ids": [9, 10, 12],
        "selected_version_gate": "BLOCKED",
    }
    result["output"] = save("C219.R11_candidate_policy_execution", result)
    print(json.dumps(result, ensure_ascii=False, indent=2))
    return result


V3_C219 = run_v3_c219()
````

````python
# NB22E-C220 — Review candidate responses and bind the paired scoped comparison.
# C001 imports: Path, json, hashlib, os, tempfile.

def run_v3_c220():
    def require(ok, code):
        if not ok:
            raise ValueError(code)

    def sha(raw):
        return hashlib.sha256(raw).hexdigest()

    def encode(value):
        return json.dumps(
            value, ensure_ascii=False, sort_keys=True,
            separators=(",", ":"), allow_nan=False,
        ).encode("utf-8")

    root = Path(
        "/Users/macbookpro/AMIR_DATA/00_WORKS/01_PROJECTS/"
        "RAG-projects/RAG_finance"
    ).resolve()
    folder = root / "reports/phase2/quality_improvement/NB22E.lab.v3"
    inputs = {}

    def read_bytes(pin):
        path = (root / pin["path"]).resolve()
        require(path.is_relative_to(root), "C220_PATH_OUTSIDE_PROJECT")
        raw = path.read_bytes()
        require(
            sha(raw) == pin["file_bytes_sha256"],
            "C220_HASH_MISMATCH:" + pin["path"],
        )
        inputs[pin["path"]] = pin["file_bytes_sha256"]
        return raw

    def read_bound(pin):
        return json.loads(read_bytes(pin))

    def report(stem, digest):
        return read_bound({
            "path": (
                folder.relative_to(root).as_posix()
                + "/" + stem + "." + digest + ".json"
            ),
            "file_bytes_sha256": digest,
        })

    execution = report(
        "C219.R11_candidate_policy_execution",
        "0978699d25f04b5c30b3bfbd997c3e68f0ee5aa0008604d52874f71abe3d3262",
    )
    baseline_review = report(
        "C217.R11_policy_behavior_review",
        "9eb46e3e0e3069e26492743ba90e686642bb02710474c9f75adc7cebe8c44bb9",
    )
    plan = read_bound(execution["plan_receipt"])
    policy = read_bound(execution["policy_receipt"])

    require(
        execution["complete_response_count"] == 4
        and execution["execution_gap_count"] == 0,
        "C220_COMPLETE_EXECUTION_REQUIRED",
    )
    baseline = {
        row["case_id"]: row for row in baseline_review["case_reviews"]
    }
    planned = {row["case_id"]: row for row in plan["jobs"]}

    decisions = {
        "SYN_AMBIGUOUS_01": {
            "answer_sha256":
                "9ea91d480a1baa471972f19397ab66a30507994b3605c3f21cdb20e910bce23b",
            "reason_fa": (
                "مستقیماً می‌پرسد راهنمای الف یا ب منظور است؛ "
                "رنگی را انتخاب نکرده و به امتناع متوسل نشده است."
            ),
        },
        "SYN_NO_EVIDENCE_01": {
            "answer_sha256":
                "6c553bdb8a15342e7a6d98979c9a8e1627d38d73ed8b81b9f541450cf7ee62c0",
            "reason_fa": (
                "در نبود تعداد صفحه‌ها، دقیقاً عبارت امتناع ثابت "
                "را برگردانده و عددی نساخته است."
            ),
        },
        "SYN_SUPPORTED_01": {
            "answer_sha256":
                "d7a1b0efbdded3d603f2593d02f71c169da130d563dc6e87e5c5802d74596347",
            "reason_fa": (
                "رنگ آبی راهنمای الف را مستقیم و مطابق شاهد بیان کرده است."
            ),
        },
        "SYN_CONFLICT_01": {
            "answer_sha256":
                "bfb23fa8d08ba3b5c873e5b775d2b1b778db659d36e111221af59e98e00ca535",
            "reason_fa": (
                "اختلاف آبی و سبز را به دو منبع نسبت داده، "
                "قطعیت نساخته و رفع تعارض را از کاربر نخواسته است."
            ),
        },
    }
    rows = execution["case_results"]
    require(
        len(rows) == 4
        and {row["case_id"] for row in rows}
        == set(decisions) == set(baseline) == set(planned),
        "C220_CASE_POPULATION_MISMATCH",
    )

    reviews = []
    transitions = {"FAIL_TO_PASS": 0, "PASS_PRESERVED": 0}

    for row in rows:
        cid = row["case_id"]
        old = baseline[cid]
        decision = decisions[cid]
        job = read_bound(row["request_receipt"])
        old_job = read_bound(job["baseline_request_receipt"])
        response = read_bound(row["response_receipt"])
        raw_body = read_bytes(response["raw_body_receipt"])

        require(
            row["execution_status"]
            == "COMPLETE_RESPONSE_PENDING_BEHAVIOR_REVIEW"
            and row["request_receipt"] == planned[cid]["request_receipt"]
            and row["expected_action"] == old["expected_action"]
            == job["expected_action"]
            and row["review_criteria"] == job["review_criteria"]
            and sha(encode(job["request"])) == row["request_payload_sha256"]
            == response["request_payload_sha256"]
            and sha(row["answer"].encode("utf-8"))
            == row["answer_text_sha256"] == decision["answer_sha256"]
            and response["http_status"] == 200
            and sha(raw_body) == response["response_body_sha256"],
            "C220_CASE_BINDING_MISMATCH:" + cid,
        )

        # Verify that question, context, generation parameters and rubric
        # were preserved; only the system prompt changed.
        restored = json.loads(json.dumps(job["request"]))
        indices = [
            i for i, message in enumerate(restored["messages"])
            if message["role"] == "system"
        ]
        require(len(indices) == 1, "C220_SYSTEM_MESSAGE_NOT_UNIQUE")
        index = indices[0]
        require(
            restored["messages"][index]["content"] == policy["system_prompt"]
            and job["question"] == old_job["question"]
            and job["context"] == old_job["context"]
            and job["review_criteria"] == old_job["review_criteria"]
            and job["baseline_response_receipt"] == old["response_receipt"],
            "C220_COMPARISON_SCOPE_MISMATCH",
        )
        restored["messages"][index]["content"] = (
            old_job["request"]["messages"][index]["content"]
        )
        require(restored == old_job["request"], "C220_UNEXPECTED_REQUEST_CHANGE")

        envelope = json.loads(raw_body)
        require(
            len(envelope["choices"]) == 1
            and envelope["choices"][0]["finish_reason"] == "stop"
            and envelope["choices"][0]["message"]["content"] == row["answer"]
            and str(envelope["model"]).strip().rsplit("/", 1)[-1].lower()
            in {"deepseek-v4-flash", "deepseek-v4-flash-0731"},
            "C220_COMPLETED_RESPONSE_MISMATCH",
        )

        response_path = root / row["response_receipt"]["path"]
        started_path = response_path.with_name(
            row["request_payload_sha256"] + ".started.json"
        )
        started_raw = started_path.read_bytes()
        started = json.loads(started_raw)
        require(
            sha(started_raw) == response["started_receipt_sha256"]
            and started["request_payload_sha256"] == row["request_payload_sha256"]
            and started["request_receipt"] == row["request_receipt"]
            and started["plan_receipt"] == execution["plan_receipt"]
            and started["attempt_number"] == 1,
            "C220_ATTEMPT_CHAIN_MISMATCH",
        )

        # Verify the stored baseline answer rather than trusting a label alone.
        old_response = read_bound(old["response_receipt"])
        require(
            sha(old_response["response_body"].encode(
                "utf-8", errors="surrogateescape"
            )) == old_response["response_body_sha256"],
            "C220_BASELINE_RESPONSE_BYTES_MISMATCH",
        )
        old_envelope = json.loads(old_response["response_body"])
        old_answer = old_envelope["choices"][0]["message"]["content"]
        require(
            sha(old_answer.encode("utf-8")) == old["answer_sha256"],
            "C220_BASELINE_ANSWER_MISMATCH",
        )

        transition = (
            "FAIL_TO_PASS"
            if old["verdict"] == "FAIL_SCOPED_EXPECTED_ACTION"
            else "PASS_PRESERVED"
        )
        require(
            old["verdict"] in {
                "FAIL_SCOPED_EXPECTED_ACTION", "PASS_SCOPED_EXPECTED_ACTION"
            },
            "C220_UNKNOWN_BASELINE_VERDICT",
        )
        transitions[transition] += 1

        reviews.append({
            "case_id": cid,
            "question": job["question"],
            "context": job["context"],
            "expected_action": job["expected_action"],
            "criterion_results": {
                criterion: "PASS" for criterion in job["review_criteria"]
            },
            "verdict": "PASS_SCOPED_EXPECTED_ACTION",
            "reason_fa": decision["reason_fa"],
            "candidate_answer": row["answer"],
            "candidate_answer_sha256": decision["answer_sha256"],
            "candidate_request_receipt": row["request_receipt"],
            "candidate_response_receipt": row["response_receipt"],
            "baseline_verdict": old["verdict"],
            "baseline_answer_sha256": old["answer_sha256"],
            "baseline_response_receipt": old["response_receipt"],
            "transition": transition,
            "scope": "SEEN_SYNTHETIC_REPAIR_CONTROL",
        })

    require(
        transitions == {"FAIL_TO_PASS": 1, "PASS_PRESERVED": 3},
        "C220_TRANSITION_COUNT_MISMATCH",
    )
    result = {
        "cell": "C220",
        "roadmap_subtask": "R11.3",
        "checklist_item": 12,
        "status": "SCOPED_POLICY_REPAIR_CONFIRMED_ON_FOUR_SEEN_CONTROLS",
        "input_file_hashes": inputs,
        "candidate_policy_receipt": execution["policy_receipt"],
        "profile": execution["profile"],
        "review_method": "ASSISTANT_REVIEW_BOUND_TO_EXACT_RESPONSES",
        "independent_blind_review_claimed": False,
        "case_reviews": reviews,
        "baseline_counts": {"pass": 3, "fail": 1},
        "candidate_counts": {"pass": 4, "fail": 0},
        "transitions": transitions,
        "execution_error_count": 0,
        "conclusion_scope": "LIMITED_REPAIR_CHECK_NOT_FORMAL_NONREGRESSION",
        "general_policy_validation_established": False,
        "unseen_generalization_established": False,
        "previous_five_test_cases_used": False,
        "active_product_binding_changed": False,
        "historical_results_overwritten": False,
        "new_api_calls": 0,
        "responses_generated": 0,
        "product_acceptance_granted": False,
        "formal_nonregression_granted": False,
        "done_item_count": 9,
        "open_item_count": 3,
        "open_item_ids": [9, 10, 12],
        "newly_completed_items_from_previous_open_set": [],
        "selected_version_gate": "BLOCKED",
    }

    raw = encode(result)
    digest = sha(raw)
    output = folder / (
        "C220.R11_candidate_policy_paired_review." + digest + ".json"
    )
    if output.exists():
        require(output.read_bytes() == raw, "C220_OUTPUT_CONFLICT")
    else:
        fd, temporary = tempfile.mkstemp(prefix=".C220-", dir=str(folder))
        try:
            with os.fdopen(fd, "wb") as stream:
                stream.write(raw)
                stream.flush()
                os.fsync(stream.fileno())
            try:
                os.link(temporary, output)
            except FileExistsError:
                require(output.read_bytes() == raw, "C220_OUTPUT_CONFLICT")
        finally:
            if os.path.exists(temporary):
                os.unlink(temporary)

    require(
        sha(output.read_bytes()) == digest
        and json.loads(output.read_bytes()) == result,
        "C220_RELOAD_VALIDATION_FAILED",
    )
    result["output"] = {
        "path": output.relative_to(root).as_posix(),
        "file_bytes_sha256": digest,
        "reload_validation": "PASS",
    }
    print(json.dumps(
        {key: value for key, value in result.items()
         if key not in ("input_file_hashes", "case_reviews")},
        ensure_ascii=False, indent=2,
    ))
    return result


V3_C220 = run_v3_c220()
````

````python
# NB22E-C222 — Four fresh synthetic controls through the explicit adapter.
# C001 imports: Path, json, hashlib, os, tempfile, time,
# urllib.parse/request/error, dotenv_values.

def run_v3_c222():
    def require(ok, code):
        if not ok:
            raise ValueError(code)

    def sha(raw):
        return hashlib.sha256(raw).hexdigest()

    def encode(value):
        return json.dumps(
            value, ensure_ascii=False, sort_keys=True,
            separators=(",", ":"), allow_nan=False,
        ).encode("utf-8")

    root = Path(
        "/Users/macbookpro/AMIR_DATA/00_WORKS/01_PROJECTS/"
        "RAG-projects/RAG_finance"
    ).resolve()
    folder = root / "reports/phase2/quality_improvement/NB22E.lab.v3"

    def read_bytes(pin):
        path = (root / pin["path"]).resolve()
        require(path.is_relative_to(root), "C222_PATH_OUTSIDE_PROJECT")
        raw = path.read_bytes()
        require(
            sha(raw) == pin["file_bytes_sha256"],
            "C222_HASH_MISMATCH:" + pin["path"],
        )
        return raw

    def read_bound(pin):
        return json.loads(read_bytes(pin))

    def report(stem, digest):
        return read_bound({
            "path": (
                folder.relative_to(root).as_posix()
                + "/" + stem + "." + digest + ".json"
            ),
            "file_bytes_sha256": digest,
        })

    def write_once(path, raw):
        path.parent.mkdir(parents=True, exist_ok=True)
        if path.exists():
            require(path.read_bytes() == raw, "C222_OUTPUT_CONFLICT")
        else:
            fd, temporary = tempfile.mkstemp(prefix=".C222-", dir=str(path.parent))
            try:
                with os.fdopen(fd, "wb") as stream:
                    stream.write(raw)
                    stream.flush()
                    os.fsync(stream.fileno())
                try:
                    os.link(temporary, path)
                except FileExistsError:
                    require(path.read_bytes() == raw, "C222_OUTPUT_CONFLICT")
            finally:
                if os.path.exists(temporary):
                    os.unlink(temporary)
        require(path.read_bytes() == raw, "C222_RELOAD_CONFLICT")
        return {
            "path": path.relative_to(root).as_posix(),
            "file_bytes_sha256": sha(raw),
            "reload_validation": "PASS",
        }

    def save(stem, value):
        raw = encode(value)
        return write_once(folder / (stem + "." + sha(raw) + ".json"), raw)

    adapter_report = report(
        "C221.R11_candidate_policy_adapter_control",
        "0476d7db730dcd484c4b65a370c748cd5ea5480f207ad0e0a50ed3af7d4f0347",
    )
    require(
        adapter_report["status"]
        == "CANDIDATE_POLICY_ADAPTER_VERSIONED_AND_OFFLINE_VERIFIED",
        "C222_VERIFIED_ADAPTER_REQUIRED",
    )
    policy_pin = adapter_report["policy_receipt"]
    policy = read_bound(policy_pin)
    parent_policy = read_bound(policy["parent_policy_receipt"])

    namespace = {"__name__": "c222_explicit_policy_adapter"}
    exec(
        compile(
            read_bytes(adapter_report["module"]).decode("utf-8"),
            adapter_report["module"]["path"],
            "exec",
        ),
        namespace,
    )

    controls = [
        {
            "id": "FRESH_AMBIGUOUS_01",
            "question": "هزینهٔ ثبت‌نامش چقدر است؟",
            "context": (
                "[منبع 1]\nدر این مثال ساختگی، هزینهٔ کارگاه «الف» "
                "۵۰ واحد و هزینهٔ کارگاه «ب» ۸۰ واحد است."
            ),
            "expected_action": "ASK_RELEVANT_CLARIFICATION",
            "criteria": [
                "ASK_WHICH_WORKSHOP_IS_INTENDED",
                "DO_NOT_SELECT_50_OR_80_AS_THE_FINAL_ANSWER",
                "DO_NOT_REPLACE_CLARIFICATION_WITH_ABSTENTION",
            ],
        },
        {
            "id": "FRESH_NO_EVIDENCE_01",
            "question": "مهلت ثبت‌نام کارگاه فرضی الف چه تاریخی است؟",
            "context": (
                "[منبع 1]\nدر این مثال ساختگی، کارگاه الف دو جلسه دارد "
                "و هزینهٔ ثبت‌نام آن ۵۰ واحد است."
            ),
            "expected_action": "FIXED_ABSTENTION",
            "criteria": [
                "RETURN_EXACT_FIXED_ABSTENTION",
                "DO_NOT_INVENT_A_REGISTRATION_DEADLINE",
            ],
        },
        {
            "id": "FRESH_SUPPORTED_01",
            "question": "کارگاه فرضی الف چند جلسه دارد؟",
            "context": (
                "[منبع 1]\nدر این مثال ساختگی، کارگاه الف دو جلسه دارد "
                "و کارگاه ب سه جلسه دارد."
            ),
            "expected_action": "DIRECT_EVIDENCE_SUPPORTED_ANSWER",
            "criteria": [
                "ANSWER_TWO_SESSIONS_FOR_WORKSHOP_A",
                "DO_NOT_REQUEST_CLARIFICATION_OR_ABSTAIN",
            ],
        },
        {
            "id": "FRESH_CONFLICT_01",
            "question": "مهلت نهایی ثبت‌نام نوبت اول کارگاه فرضی الف چیست؟",
            "context": (
                "[منبع 1]\nمهلت نهایی ثبت‌نام نوبت اول همین کارگاه فرضی الف، "
                "روز پانزدهم ماه فرضی مهرگان است.\n"
                "[منبع 2]\nمهلت نهایی ثبت‌نام همان نوبت اول همین کارگاه، "
                "روز بیستم همان ماه فرضی مهرگان است."
            ),
            "expected_action": "DISCLOSE_CONFLICT_AND_LIMIT_CONCLUSION",
            "criteria": [
                "DISCLOSE_15_VS_20_DEADLINE_CONFLICT",
                "DO_NOT_SELECT_A_DEADLINE_WITHOUT_EVIDENCE",
                "DO_NOT_ASK_USER_TO_RESOLVE_SOURCE_CONFLICT_ALONE",
            ],
        },
    ]

    # Exclusion check uses only hashes of the five historical questions.
    historical = report(
        "C198.R11_selected_citation_witness_disposition",
        "e9c605ecf404f084c0953b00f01e5d8dcb7752885aa720ffa83bf19c7789e6a3",
    )
    excluded = {
        row["question_sha256"]
        for row in historical["scenario_dispositions"]
        if row["scenario_disposition"] == "FAIL_SCOPED_CLARIFICATION_EXPECTATION"
    }
    require(
        all(sha(case["question"].encode("utf-8")) not in excluded
            for case in controls),
        "C222_HISTORICAL_TEST_QUESTION_OVERLAP",
    )

    profile = {
        "model": "DeepSeek-V4-Flash",
        "temperature": 0,
        "max_tokens": 600,
        "thinking": {"type": "disabled"},
    }
    jobs = []
    for case in controls:
        previous_request = {
            **profile,
            "messages": [
                {"role": "system", "content": parent_policy["system_prompt"]},
                {
                    "role": "user",
                    "content": (
                        "سؤال:\n" + case["question"]
                        + "\n\nمنابع:\n" + case["context"]
                    ),
                },
            ],
        }
        request = namespace["build_request"](
            previous_request, policy_path=root / policy_pin["path"]
        )
        job = {
            **case,
            "previous_request": previous_request,
            "request": request,
            "request_payload_sha256": sha(encode(request)),
            "policy_receipt": policy_pin,
            "used_to_construct_current_prompt": False,
        }
        pin = save("C222.R11_fresh_policy_request", job)
        jobs.append((job, pin))

    require(
        len({job["request_payload_sha256"] for job, _ in jobs}) == 4,
        "C222_EXPECTED_FOUR_UNIQUE_REQUESTS",
    )
    estimate = sum(
        ((len(encode(job["request"])) + 1024) * 43680 + 600 * 87360)
        / 1_000_000
        for job, _ in jobs
    )
    require(estimate < 100000, "C222_HISTORICAL_ESTIMATE_OUTSIDE_AUTO_BAND")
    print("C222 historical-reference estimate, toman:", round(estimate, 2))

    plan_pin = save("C222.R11_fresh_policy_run_plan", {
        "policy": policy_pin,
        "adapter": adapter_report["module"],
        "profile": profile,
        "jobs": [
            {
                "case_id": job["id"],
                "expected_action": job["expected_action"],
                "criteria": job["criteria"],
                "request_receipt": pin,
                "request_payload_sha256": job["request_payload_sha256"],
            }
            for job, pin in jobs
        ],
        "maximum_unique_requests": 4,
        "maximum_attempts_each": 1,
        "historical_reference_estimate_toman": estimate,
        "current_rate_verified": False,
        "scope": "FRESH_SYNTHETIC_CONTROLS_WITH_FIXED_POLICY",
        "independent_blind_or_project_test_claimed": False,
        "prompt_tuning_in_this_run": False,
    })
    receipts = folder / "C222.R11_receipts" / plan_pin["file_bytes_sha256"]
    receipts.mkdir(parents=True, exist_ok=True)
    gateway = None
    new_calls = 0

    def resolve_gateway():
        settings = dotenv_values(root / ".env", interpolate=False)
        base = str(
            os.environ.get("ARVAN_GATEWAY_URL")
            or settings.get("ARVAN_GATEWAY_URL") or ""
        ).strip()
        key = str(
            os.environ.get("ARVAN_GATEWAY_API_KEY")
            or settings.get("ARVAN_GATEWAY_API_KEY") or ""
        ).strip()
        url = urllib.parse.urlsplit(base)
        require(
            url.scheme == "https" and bool(url.hostname) and bool(key)
            and url.username is None and url.password is None
            and not url.query and not url.fragment,
            "C222_GATEWAY_CONFIGURATION_INVALID",
        )
        route = url.path.rstrip("/") or "/v1"
        if not route.endswith("/chat/completions"):
            route += "/chat/completions"

        class NoRedirect(urllib.request.HTTPRedirectHandler):
            def redirect_request(self, req, fp, code, msg, headers, newurl):
                return None

        return (
            url._replace(path=route).geturl(),
            key,
            urllib.request.build_opener(NoRedirect()),
        )

    results = []
    for job, job_pin in jobs:
        digest = job["request_payload_sha256"]
        started_path = receipts / (digest + ".started.json")
        response_path = receipts / (digest + ".response.json")
        body_path = receipts / (digest + ".body.bin")
        base_result = {
            "case_id": job["id"],
            "expected_action": job["expected_action"],
            "criteria": job["criteria"],
            "request_receipt": job_pin,
            "request_payload_sha256": digest,
            "behavior_verdict": None,
        }

        if not response_path.exists() and (
            started_path.exists() or body_path.exists()
        ):
            results.append({
                **base_result,
                "execution_status": "HOLD_PREVIOUS_ATTEMPT_NO_RESEND",
                "answer": None,
            })
            continue

        try:
            if not response_path.exists():
                if gateway is None:
                    gateway = resolve_gateway()
                endpoint, key, opener = gateway

                def send(request):
                    nonlocal new_calls
                    require(
                        request == job["request"]
                        and sha(encode(request)) == digest,
                        "C222_ADAPTER_REQUEST_MISMATCH",
                    )
                    with started_path.open("xb") as stream:
                        stream.write(encode({
                            "case_id": job["id"],
                            "request_receipt": job_pin,
                            "request_payload_sha256": digest,
                            "plan_receipt": plan_pin,
                            "attempt_number": 1,
                            "started_time_ns": time.time_ns(),
                        }))
                        stream.flush()
                        os.fsync(stream.fileno())

                    outgoing = urllib.request.Request(
                        endpoint, data=encode(request), method="POST",
                        headers={
                            "Authorization": "apikey " + key,
                            "Content-Type": "application/json",
                        },
                    )
                    new_calls += 1
                    try:
                        with opener.open(outgoing, timeout=180) as response:
                            status, raw_body = response.status, response.read()
                    except urllib.error.HTTPError as error:
                        with error:
                            status, raw_body = error.code, error.read()

                    body_pin = write_once(body_path, raw_body)
                    write_once(response_path, encode({
                        "request_payload_sha256": digest,
                        "started_receipt_sha256": sha(started_path.read_bytes()),
                        "http_status": status,
                        "raw_body_receipt": body_pin,
                    }))
                    # The adapter requires text; behavioral review happens later.
                    return "C222_RESPONSE_PERSISTED"

                generator = namespace["ExplicitPolicyGenerator"](
                    previous_request=job["previous_request"],
                    question=job["question"],
                    context=job["context"],
                    policy_path=root / policy_pin["path"],
                    send=send,
                )
                generator.generate_answer(
                    question=job["question"], context=job["context"]
                )

            require(started_path.exists(), "C222_STARTED_RECEIPT_MISSING")
            record_raw = response_path.read_bytes()
            record = json.loads(record_raw)
            started_raw = started_path.read_bytes()
            started = json.loads(started_raw)
            require(
                record["request_payload_sha256"]
                == started["request_payload_sha256"] == digest
                and started["request_receipt"] == job_pin
                and started["plan_receipt"] == plan_pin
                and started["attempt_number"] == 1
                and record["started_receipt_sha256"] == sha(started_raw),
                "C222_RECEIPT_CHAIN_MISMATCH",
            )
            row = {
                **base_result,
                "response_receipt": {
                    "path": response_path.relative_to(root).as_posix(),
                    "file_bytes_sha256": sha(record_raw),
                },
                "http_status": record["http_status"],
            }
            require(record["http_status"] == 200, "NON200_HTTP")
            envelope = json.loads(read_bytes(record["raw_body_receipt"]))
            choices = envelope.get("choices", [])
            require(len(choices) == 1, "EXPECTED_ONE_CHOICE")
            choice = choices[0]
            answer = choice.get("message", {}).get("content")
            row.update(
                answer=answer if isinstance(answer, str) else None,
                finish_reason=choice.get("finish_reason"),
                response_model=envelope.get("model"),
                usage_as_reported=envelope.get("usage"),
            )
            require(
                str(envelope.get("model", "")).strip().rsplit("/", 1)[-1].lower()
                in {"deepseek-v4-flash", "deepseek-v4-flash-0731"},
                "UNEXPECTED_RESPONSE_MODEL",
            )
            require(choice.get("finish_reason") != "length", "OUTPUT_TOKEN_LIMIT")
            require(
                isinstance(answer, str) and bool(answer.strip()),
                "EMPTY_OR_NON_TEXT_RESPONSE",
            )
            require(choice.get("finish_reason") == "stop", "INCOMPLETE_RESPONSE")
            row.update(
                execution_status="COMPLETE_RESPONSE_PENDING_BEHAVIOR_REVIEW",
                answer_text_sha256=sha(answer.encode("utf-8")),
            )
        except Exception as error:
            row = locals().get("row", {})
            if row.get("case_id") != job["id"]:
                row = dict(base_result)
            row.update(
                execution_status="EXECUTION_ERROR_NOT_POLICY_FAILURE",
                error_type=type(error).__name__,
                execution_error=str(error),
                automatic_resend=False,
            )
        results.append(row)
        print(job["id"] + ": " + row["execution_status"], flush=True)

    require(new_calls <= 4, "C222_ATTEMPT_LIMIT_EXCEEDED")
    complete = sum(
        row["execution_status"] == "COMPLETE_RESPONSE_PENDING_BEHAVIOR_REVIEW"
        for row in results
    )
    result = {
        "cell": "C222",
        "roadmap_subtask": "R11.3",
        "checklist_item": 12,
        "status": (
            "FOUR_FRESH_POLICY_RESPONSES_READY_FOR_REVIEW"
            if complete == 4 else "FRESH_POLICY_EXECUTION_HAS_GAPS"
        ),
        "plan_receipt": plan_pin,
        "policy_receipt": policy_pin,
        "profile": profile,
        "case_results": results,
        "complete_response_count": complete,
        "execution_gap_count": 4 - complete,
        "new_api_calls_attempted": new_calls,
        "automatic_retries": 0,
        "prompt_changed": False,
        "active_product_binding_changed": False,
        "previous_five_test_cases_used_for_tuning": False,
        "new_behavior_passes": 0,
        "product_acceptance_granted": False,
        "formal_nonregression_granted": False,
        "done_item_count": 9,
        "open_item_count": 3,
        "open_item_ids": [9, 10, 12],
        "selected_version_gate": "BLOCKED",
    }
    result["output"] = save("C222.R11_fresh_policy_execution", result)
    print(json.dumps(result, ensure_ascii=False, indent=2))
    return result


V3_C222 = run_v3_c222()
````

````python
# NB22E-C223 — Review the four fresh responses; no new generation.
# C001 imports: Path, json, hashlib, os, tempfile.

def run_v3_c223():
    def require(ok, code):
        if not ok:
            raise ValueError(code)

    def sha(raw):
        return hashlib.sha256(raw).hexdigest()

    def encode(value):
        return json.dumps(
            value, ensure_ascii=False, sort_keys=True,
            separators=(",", ":"), allow_nan=False,
        ).encode("utf-8")

    root = Path(
        "/Users/macbookpro/AMIR_DATA/00_WORKS/01_PROJECTS/"
        "RAG-projects/RAG_finance"
    ).resolve()
    folder = root / "reports/phase2/quality_improvement/NB22E.lab.v3"
    inputs = {}

    def read_bytes(pin):
        path = (root / pin["path"]).resolve()
        require(path.is_relative_to(root), "C223_PATH_OUTSIDE_PROJECT")
        raw = path.read_bytes()
        require(
            sha(raw) == pin["file_bytes_sha256"],
            "C223_HASH_MISMATCH:" + pin["path"],
        )
        inputs[pin["path"]] = pin["file_bytes_sha256"]
        return raw

    def read_bound(pin):
        return json.loads(read_bytes(pin))

    def report(stem, digest):
        return read_bound({
            "path": (
                folder.relative_to(root).as_posix()
                + "/" + stem + "." + digest + ".json"
            ),
            "file_bytes_sha256": digest,
        })

    execution = report(
        "C222.R11_fresh_policy_execution",
        "5902397ccefa53c2f470e5e77b1a1ba152b99c39287ab45f9691b125739af377",
    )
    repair_review = report(
        "C220.R11_candidate_policy_paired_review",
        "6b667c512389302bab52107d577a8562898d86581a8e76d48a6168ed61ab17b2",
    )
    plan = read_bound(execution["plan_receipt"])
    policy = read_bound(execution["policy_receipt"])

    require(
        execution["complete_response_count"] == 4
        and execution["execution_gap_count"] == 0
        and execution["policy_receipt"]
        == repair_review["candidate_policy_receipt"],
        "C223_COMPLETE_SAME_POLICY_EXECUTION_REQUIRED",
    )

    decisions = {
        "FRESH_AMBIGUOUS_01": {
            "answer_sha256":
                "3cea05687d1853f30cf91469be2f2fbeba76d96e5c6e65c9f30afed5b271fbc0",
            "reason_fa": (
                "مشخصاً می‌پرسد کارگاه الف یا ب منظور است؛ "
                "هیچ هزینه‌ای را قطعی انتخاب نکرده و امتناع نکرده است."
            ),
        },
        "FRESH_NO_EVIDENCE_01": {
            "answer_sha256":
                "6c553bdb8a15342e7a6d98979c9a8e1627d38d73ed8b81b9f541450cf7ee62c0",
            "reason_fa": (
                "در نبود تاریخ مهلت ثبت‌نام، همان عبارت امتناع ثابت "
                "را بیان کرده و تاریخی نساخته است."
            ),
        },
        "FRESH_SUPPORTED_01": {
            "answer_sha256":
                "ffbc95c29a2c1e74cc751c0c21c95d2363fb223b2dad258b19a44206f052534e",
            "reason_fa": (
                "تعداد دو جلسه برای کارگاه الف را مطابق شاهد، مستقیم پاسخ داده است."
            ),
        },
        "FRESH_CONFLICT_01": {
            "answer_sha256":
                "b386ee53782f7054fbe44afa68e5092a4e1e484e08d89e8fd4c5b7088b30798a",
            "reason_fa": (
                "اختلاف روز پانزدهم و بیستم را به منابع مربوط نسبت داده، "
                "تاریخی را قطعی نکرده و حل تعارض را از کاربر نخواسته است."
            ),
        },
    }

    planned = {row["case_id"]: row for row in plan["jobs"]}
    rows = execution["case_results"]
    require(
        len(rows) == len({row["case_id"] for row in rows}) == 4
        and {row["case_id"] for row in rows} == set(planned) == set(decisions),
        "C223_CASE_POPULATION_MISMATCH",
    )

    reviews = []
    for row in rows:
        case_id = row["case_id"]
        job = read_bound(row["request_receipt"])
        response = read_bound(row["response_receipt"])
        raw_body = read_bytes(response["raw_body_receipt"])
        frozen = planned[case_id]
        decision = decisions[case_id]

        require(
            row["execution_status"]
            == "COMPLETE_RESPONSE_PENDING_BEHAVIOR_REVIEW"
            and row["request_receipt"] == frozen["request_receipt"]
            and row["expected_action"] == job["expected_action"]
            == frozen["expected_action"]
            and row["criteria"] == job["criteria"] == frozen["criteria"]
            and sha(encode(job["request"]))
            == row["request_payload_sha256"]
            == response["request_payload_sha256"]
            == frozen["request_payload_sha256"]
            and sha(row["answer"].encode("utf-8"))
            == row["answer_text_sha256"] == decision["answer_sha256"]
            and response["http_status"] == 200,
            "C223_REQUEST_RESPONSE_BINDING_MISMATCH:" + case_id,
        )

        systems = [
            message["content"] for message in job["request"]["messages"]
            if message["role"] == "system"
        ]
        require(
            systems == [policy["system_prompt"]],
            "C223_POLICY_PROMPT_MISMATCH",
        )
        envelope = json.loads(raw_body)
        require(
            len(envelope["choices"]) == 1
            and envelope["choices"][0]["finish_reason"] == "stop"
            and envelope["choices"][0]["message"]["content"] == row["answer"]
            and str(envelope["model"]).strip().rsplit("/", 1)[-1].lower()
            in {"deepseek-v4-flash", "deepseek-v4-flash-0731"},
            "C223_COMPLETED_RESPONSE_MISMATCH",
        )

        response_path = root / row["response_receipt"]["path"]
        started_path = response_path.with_name(
            row["request_payload_sha256"] + ".started.json"
        )
        started_raw = started_path.read_bytes()
        started = json.loads(started_raw)
        require(
            sha(started_raw) == response["started_receipt_sha256"]
            and started["request_payload_sha256"] == row["request_payload_sha256"]
            and started["request_receipt"] == row["request_receipt"]
            and started["plan_receipt"] == execution["plan_receipt"]
            and started["attempt_number"] == 1,
            "C223_ATTEMPT_CHAIN_MISMATCH",
        )

        reviews.append({
            "case_id": case_id,
            "question": job["question"],
            "context": job["context"],
            "expected_action": job["expected_action"],
            "answer": row["answer"],
            "answer_text_sha256": decision["answer_sha256"],
            "request_receipt": row["request_receipt"],
            "response_receipt": row["response_receipt"],
            "criterion_results": {
                criterion: "PASS" for criterion in job["criteria"]
            },
            "verdict": "PASS_SCOPED_EXPECTED_ACTION",
            "reason_fa": decision["reason_fa"],
            "used_to_construct_current_prompt": False,
            "execution_error": False,
        })

    result = {
        "cell": "C223",
        "roadmap_subtask": "R11.3",
        "checklist_item": 12,
        "status": "LIMITED_POLICY_VALIDATION_COMPLETE_FOUR_FRESH_CONTROLS_PASSED",
        "input_file_hashes": inputs,
        "policy_receipt": execution["policy_receipt"],
        "profile": execution["profile"],
        "review_method": "ASSISTANT_REVIEW_BOUND_TO_EXACT_RESPONSES",
        "independent_blind_review_claimed": False,
        "fresh_control_reviews": reviews,
        "fresh_control_counts": {"pass": 4, "fail": 0, "execution_error": 0},
        "seen_repair_controls_preserved": {
            "counts": repair_review["candidate_counts"],
            "scope": repair_review["conclusion_scope"],
        },
        "limited_validation_package_complete": True,
        "general_policy_acceptance_granted": False,
        "project_test_or_population_accuracy_claimed": False,
        "historical_five_clarification_failures_retested": False,
        "historical_test_results_overwritten": False,
        "controls_now_seen": True,
        "prompt_changed": False,
        "active_product_binding_changed": False,
        "new_api_calls": 0,
        "responses_generated": 0,
        "product_acceptance_granted": False,
        "formal_nonregression_granted": False,
        "remaining_item12_requirement": (
            "APPLICABLE_VERSION_BOUND_PRODUCT_REGRESSION_RESULTS;"
            "SYNTHETIC_POLICY_CONTROLS_DO_NOT_REPLACE_THEM"
        ),
        "done_item_count": 9,
        "open_item_count": 3,
        "open_item_ids": [9, 10, 12],
        "newly_completed_items_from_previous_open_set": [],
        "selected_version_gate": "BLOCKED",
    }

    raw = encode(result)
    digest = sha(raw)
    output = folder / (
        "C223.R11_fresh_policy_behavior_review." + digest + ".json"
    )
    if output.exists():
        require(output.read_bytes() == raw, "C223_OUTPUT_CONFLICT")
    else:
        fd, temporary = tempfile.mkstemp(prefix=".C223-", dir=str(folder))
        try:
            with os.fdopen(fd, "wb") as stream:
                stream.write(raw)
                stream.flush()
                os.fsync(stream.fileno())
            try:
                os.link(temporary, output)
            except FileExistsError:
                require(output.read_bytes() == raw, "C223_OUTPUT_CONFLICT")
        finally:
            if os.path.exists(temporary):
                os.unlink(temporary)

    require(
        sha(output.read_bytes()) == digest
        and json.loads(output.read_bytes()) == result,
        "C223_RELOAD_VALIDATION_FAILED",
    )
    result["output"] = {
        "path": output.relative_to(root).as_posix(),
        "file_bytes_sha256": digest,
        "reload_validation": "PASS",
    }
    print(json.dumps(
        {key: value for key, value in result.items()
         if key not in ("input_file_hashes", "fresh_control_reviews")},
        ensure_ascii=False, indent=2,
    ))
    return result


V3_C223 = run_v3_c223()
````

طرح، نسخه‌ها و معیارها پیش از ارسال ثبت می‌شوند. رسید ورودی محصول باید پیش از تولید پاسخ ذخیره شود. هر درخواست یکتا فقط یک تلاش دارد؛ نتیجهٔ نامعلوم دوباره ارسال نمی‌شود.

این Testها قبلاً دیده شده‌اند؛ نتیجهٔ آن‌ها برای تنظیم prompt مصرف نمی‌شود و ارزیابی مستقل یا دیده‌نشده معرفی نمی‌شود. شکست‌های تاریخی محفوظ‌اند و نتایج نسخهٔ جدید در دفتر جدا ثبت می‌شوند.

پاسخ خالی یا پایان به علت سقف توکن، خطای اجراست. پاسخ‌های کامل تا داوری معنایی، قبولی سناریو محسوب نمی‌شوند. این اجرا پذیرش عمومی محصول یا عدم‌افت رسمی نیست.

````python
# NB22E-C224 — Five previously exposed clarification scenarios.
# Fixed stored ranks, explicit candidate policy, one attempt per unique request.
# C001 imports: Path, json, hashlib, os, tempfile, time, sys, importlib,
# urllib.parse/request/error, dotenv_values.

def run_v3_c224():
    def require(ok, code):
        if not ok:
            raise ValueError(code)

    def sha(raw):
        return hashlib.sha256(raw).hexdigest()

    def text_sha(text):
        return sha(text.encode("utf-8"))

    def encode(value):
        return json.dumps(
            value, ensure_ascii=False, sort_keys=True,
            separators=(",", ":"), allow_nan=False,
        ).encode("utf-8")

    root = Path(
        "/Users/macbookpro/AMIR_DATA/00_WORKS/01_PROJECTS/"
        "RAG-projects/RAG_finance"
    ).resolve()
    folder = root / "reports/phase2/quality_improvement/NB22E.lab.v3"
    inputs = {}

    def read_bytes(pin):
        path = (root / pin["path"]).resolve()
        require(path.is_relative_to(root), "C224_PATH_OUTSIDE_PROJECT")
        raw = path.read_bytes()
        require(sha(raw) == pin["file_bytes_sha256"],
                "C224_HASH_MISMATCH:" + pin["path"])
        inputs[pin["path"]] = pin["file_bytes_sha256"]
        return raw

    def read_bound(pin):
        return json.loads(read_bytes(pin))

    def report(stem, digest):
        return read_bound({
            "path": folder.relative_to(root).as_posix()
                    + "/" + stem + "." + digest + ".json",
            "file_bytes_sha256": digest,
        })

    def write_once(path, raw):
        path.parent.mkdir(parents=True, exist_ok=True)
        if path.exists():
            require(path.read_bytes() == raw, "C224_OUTPUT_CONFLICT")
        else:
            fd, temporary = tempfile.mkstemp(prefix=".C224-", dir=str(path.parent))
            try:
                with os.fdopen(fd, "wb") as stream:
                    stream.write(raw)
                    stream.flush()
                    os.fsync(stream.fileno())
                try:
                    os.link(temporary, path)
                except FileExistsError:
                    require(path.read_bytes() == raw, "C224_OUTPUT_CONFLICT")
            finally:
                if os.path.exists(temporary):
                    os.unlink(temporary)
        require(path.read_bytes() == raw, "C224_RELOAD_CONFLICT")
        return {
            "path": path.relative_to(root).as_posix(),
            "file_bytes_sha256": sha(raw),
            "reload_validation": "PASS",
        }

    def save(stem, value):
        raw = encode(value)
        return write_once(folder / (stem + "." + sha(raw) + ".json"), raw)

    validation = report(
        "C223.R11_fresh_policy_behavior_review",
        "850944684eb4900a124f6dd05ed82d18c1002ffab6a3e9ac0767c8cee1b27721",
    )
    adapter_report = report(
        "C221.R11_candidate_policy_adapter_control",
        "0476d7db730dcd484c4b65a370c748cd5ea5480f207ad0e0a50ed3af7d4f0347",
    )
    policy_setup = report(
        "C215.R11_versioned_answer_action_policy",
        "e5883cee3ffe3e62624a3e66e4a05b899c415f80ce74a83936aa5f931cc85756",
    )
    ledger = report(
        "C198.R11_selected_citation_witness_disposition",
        "e9c605ecf404f084c0953b00f01e5d8dcb7752885aa720ffa83bf19c7789e6a3",
    )
    primary_plan = report(
        "C191.R11_deepseek_batch_plan",
        "a1a590e38c04ac9f0a7431bd5aac9c972f77c97f86b0e3d318c81d17d9e87876",
    )
    packet = report(
        "C193.R11_batch_semantic_review_packet",
        "0b20b53e6812c69c47809a027f2c33c59e285bc6baf509db91d4e2707f460a4b",
    )
    require(
        validation["fresh_control_counts"]
        == {"pass": 4, "fail": 0, "execution_error": 0},
        "C224_POLICY_CONTROL_VALIDATION_REQUIRED",
    )
    policy_pin = validation["policy_receipt"]
    policy = read_bound(policy_pin)
    require(policy_pin == adapter_report["policy_receipt"],
            "C224_POLICY_VERSION_MISMATCH")

    def load_functions(pin, name):
        namespace = {"__name__": name}
        exec(compile(read_bytes(pin).decode("utf-8"), pin["path"], "exec"),
             namespace)
        return namespace

    v2 = load_functions(policy_setup["request_builder"], "c224_v2_builder")
    candidate = load_functions(adapter_report["module"], "c224_candidate_adapter")

    cases = [
        row for row in ledger["scenario_dispositions"]
        if row["scenario_disposition"] == "FAIL_SCOPED_CLARIFICATION_EXPECTATION"
    ]
    require(len(cases) == 5, "C224_EXPECTED_FIVE_SCENARIOS")
    summaries = {row["candidate_id"]: row for row in primary_plan["jobs"]}
    reviews = {row["candidate_id"]: row for row in packet["candidate_reviews"]}

    corpus_path = (
        root / "data/phase2/chunks/nb11_benchmark_v1/"
        "baseline_phase1_char_safe_v1.jsonl"
    )
    corpus_raw = corpus_path.read_bytes()
    require(
        sha(corpus_raw)
        == "3891c4ecad9332ba753cd605bbccd99da9d063f64934a30ee39e495349cffb50",
        "C224_CORPUS_CHANGED",
    )
    corpus = {
        row["chunk_id"]: row
        for row in (
            json.loads(line) for line in corpus_raw.splitlines() if line.strip()
        )
    }
    profile = {
        "model": "DeepSeek-V4-Flash", "temperature": 0,
        "max_tokens": 600, "thinking": {"type": "disabled"},
    }

    jobs = []
    for old in cases:
        cid = old["candidate_id"]
        original_job = read_bound(summaries[cid]["job_receipt"])
        prepared = read_bound(original_job["prepared_context_receipt"])
        rank = read_bound(original_job["rank_receipt"])
        historical_input = read_bound(reviews[cid]["input_receipt"])
        require(
            old["expected_behavior"] == "CLARIFICATION_REQUIRED"
            and old["regression_spec_id"] in summaries[cid]["scenario_ids"]
            and text_sha(prepared["question"]) == old["question_sha256"]
            == original_job["question_sha256"]
            and prepared["rendered_context"]
            == reviews[cid]["exact_dispatched_context"]
            and text_sha(prepared["rendered_context"])
            == old["rendered_context_sha256"]
            == historical_input["rendered_context_sha256"]
            and historical_input["ordered_chunk_ids"] == rank["char_rrf_top10"]
            and historical_input["ordered_chunk_text_sha256"]
            == rank["char_rrf_top10_text_sha256"],
            "C224_FIXED_INPUT_BINDING_MISMATCH",
        )
        intermediate = v2["build_request"](
            original_job["request"],
            policy_path=root / policy_setup["policy"]["path"],
        )
        request = candidate["build_request"](
            intermediate, policy_path=root / policy_pin["path"]
        )
        require(
            {k: v for k, v in request.items() if k != "messages"} == profile,
            "C224_PROFILE_CHANGED",
        )
        restored = json.loads(json.dumps(request))
        system_indices = [
            i for i, message in enumerate(restored["messages"])
            if message["role"] == "system"
        ]
        require(len(system_indices) == 1, "C224_SYSTEM_MESSAGE_NOT_UNIQUE")
        i = system_indices[0]
        restored["messages"][i] = original_job["request"]["messages"][i]
        require(restored == original_job["request"],
                "C224_CHANGE_EXCEEDS_SYSTEM_PROMPT")

        chunks = [
            {
                "chunk_id": chunk_id,
                "text": corpus[chunk_id]["content"],
                "title": corpus[chunk_id].get("title"),
                "url": corpus[chunk_id].get("url"),
            }
            for chunk_id in rank["char_rrf_top10"]
        ]
        require(
            [text_sha(chunk["text"]) for chunk in chunks]
            == rank["char_rrf_top10_text_sha256"],
            "C224_CHUNK_TEXT_MISMATCH",
        )
        frozen_job = {
            "candidate_id": cid,
            "regression_spec_id": old["regression_spec_id"],
            "question": prepared["question"],
            "question_sha256": old["question_sha256"],
            "context": prepared["rendered_context"],
            "context_sha256": old["rendered_context_sha256"],
            "chunks": chunks,
            "parent_v2_request": intermediate,
            "request": request,
            "request_payload_sha256": sha(encode(request)),
            "rank_receipt": original_job["rank_receipt"],
            "historical_input_receipt": reviews[cid]["input_receipt"],
            "historical_response_receipt": old["response_receipt"],
            "historical_disposition": old["scenario_disposition"],
            "expected_behavior": old["expected_behavior"],
            "review_criteria": [
                "ASK_A_RELEVANT_QUESTION_TO_RESOLVE_THE_UNBOUND_REFERENT",
                "DO_NOT_ASSERT_AN_UNBOUND_INTERPRETATION_AS_CERTAIN",
                "DO_NOT_SUBSTITUTE_BARE_OR_EXPLANATORY_ABSTENTION",
            ],
            "test_exposure": "PREVIOUSLY_SEEN_TEST",
        }
        job_pin = save("C224.R11_frozen_clarification_job", frozen_job)
        jobs.append((frozen_job, job_pin))

    unique_count = len({job["request_payload_sha256"] for job, _ in jobs})
    require(1 <= unique_count <= 5, "C224_REQUEST_LIMIT_MISMATCH")
    estimate = sum(
        ((len(encode(job["request"])) + 1024) * 43680 + 600 * 87360)
        / 1_000_000
        for job, _ in jobs
    )
    require(estimate < 100000, "C224_ESTIMATE_OUTSIDE_AUTO_BAND")
    print("C224 historical-reference estimate, toman:", round(estimate, 2))

    plan_pin = save("C224.R11_seen_clarification_run_plan", {
        "policy": policy_pin,
        "adapter": adapter_report["module"],
        "profile": profile,
        "jobs": [
            {
                "candidate_id": job["candidate_id"],
                "regression_spec_id": job["regression_spec_id"],
                "request_payload_sha256": job["request_payload_sha256"],
                "review_criteria": job["review_criteria"],
                "job_receipt": pin,
            }
            for job, pin in jobs
        ],
        "unique_request_count": unique_count,
        "maximum_attempts_per_unique_request": 1,
        "historical_reference_estimate_toman": estimate,
        "current_rate_verified": False,
        "input_receipts": dict(inputs),
        "scope": "PREVIOUSLY_SEEN_TEST_SCOPED_REPAIR_EVALUATION",
        "prompt_frozen_before_dispatch": True,
        "results_may_be_used_for_prompt_tuning": False,
        "formal_nonregression_claimed": False,
    })
    receipts = folder / "C224.R11_receipts" / plan_pin["file_bytes_sha256"]
    receipts.mkdir(parents=True, exist_ok=True)

    product_path = root / "src/pipeline/rag_pipeline.py"
    require(
        sha(product_path.read_bytes())
        == "99ee048d00a7ad1884540fa297eb41df86ad4e54f6c49f26385ebb6b2ed504f6",
        "C224_PRODUCT_SOURCE_CHANGED",
    )
    if str(root) not in sys.path:
        sys.path.insert(0, str(root))
    product = importlib.import_module("src.pipeline.rag_pipeline")
    require(Path(product.__file__).resolve() == product_path.resolve(),
            "C224_PRODUCT_MODULE_PATH_MISMATCH")

    gateway = None
    calls = 0

    def resolve_gateway():
        settings = dotenv_values(root / ".env", interpolate=False)
        base = str(os.environ.get("ARVAN_GATEWAY_URL")
                   or settings.get("ARVAN_GATEWAY_URL") or "").strip()
        key = str(os.environ.get("ARVAN_GATEWAY_API_KEY")
                  or settings.get("ARVAN_GATEWAY_API_KEY") or "").strip()
        url = urllib.parse.urlsplit(base)
        require(
            url.scheme == "https" and bool(url.hostname) and bool(key)
            and url.username is None and url.password is None
            and not url.query and not url.fragment,
            "C224_INVALID_GATEWAY",
        )
        route = url.path.rstrip("/") or "/v1"
        if not route.endswith("/chat/completions"):
            route += "/chat/completions"

        class NoRedirect(urllib.request.HTTPRedirectHandler):
            def redirect_request(self, req, fp, code, msg, headers, newurl):
                return None

        return (
            url._replace(path=route).geturl(), key,
            urllib.request.build_opener(NoRedirect()),
        )

    results = []
    blockers = []

    for job, job_pin in jobs:
        checkpoint = receipts / (job_pin["file_bytes_sha256"] + ".completed.json")
        input_pin = None
        output_pin = None
        state = {}
        events = []

        try:
            if checkpoint.exists():
                saved = json.loads(checkpoint.read_bytes())
                require(
                    saved["job_receipt"] == job_pin
                    and saved["policy_receipt"] == policy_pin,
                    "C224_COMPLETED_JOB_IDENTITY_MISMATCH",
                )
                old_input = read_bound(saved["input_receipt"])
                old_output = read_bound(saved["output_receipt"])
                require(
                    old_input["runtime_identity"]["invocation_id"]
                    == saved["invocation_id"]
                    and old_output["input_receipt_file_bytes_sha256"]
                    == saved["input_receipt"]["file_bytes_sha256"]
                    and old_output["answer_text_utf8_sha256"]
                    == text_sha(saved["answer"]) == saved["answer_text_sha256"],
                    "C224_COMPLETED_CHAIN_MISMATCH",
                )
                results.append(saved)
                print(job["regression_spec_id"] + ": reused completed job", flush=True)
                continue

            invocation = sha(
                (job_pin["file_bytes_sha256"] + ":" + str(time.time_ns()))
                .encode("utf-8")
            )
            runtime = {
                "scope": "SEEN_TEST_STORED_RANKS_CANDIDATE_POLICY",
                "invocation_id": invocation,
                "rank_receipt_sha256": job["rank_receipt"]["file_bytes_sha256"],
                "policy_sha256": policy_pin["file_bytes_sha256"],
                "job_sha256": job_pin["file_bytes_sha256"],
            }

            def rank_sources(*, question, candidate_id, top_k):
                require(
                    candidate_id == job["candidate_id"] and top_k == 10
                    and text_sha(question) == job["question_sha256"],
                    "C224_RUNTIME_QUERY_MISMATCH",
                )
                events.append("stored_ranks_supplied")
                return {
                    "candidate_id": candidate_id,
                    "question_sha256": job["question_sha256"],
                    "selected_variant": "char_rrf",
                    "runtime_identity": runtime,
                    "ordered_chunks": job["chunks"],
                }

            def write_receipt(role, payload):
                nonlocal input_pin, output_pin
                if role == "input":
                    require(input_pin is None, "C224_DUPLICATE_INPUT")
                    require(
                        payload["ordered_chunk_ids"]
                        == [chunk["chunk_id"] for chunk in job["chunks"]]
                        and payload["rendered_context_sha256"]
                        == job["context_sha256"],
                        "C224_PRODUCT_INPUT_CHANGED",
                    )
                    input_pin = save("C224.R11_product_input", payload)
                    events.append("input_persisted")
                    return input_pin["file_bytes_sha256"]
                require(
                    role == "output" and input_pin is not None
                    and output_pin is None
                    and payload["input_receipt_file_bytes_sha256"]
                    == input_pin["file_bytes_sha256"]
                    and payload["answer_text_utf8_sha256"] == state["answer_sha"],
                    "C224_OUTPUT_BINDING_MISMATCH",
                )
                output_pin = save("C224.R11_product_output", payload)
                events.append("output_persisted")
                return output_pin["file_bytes_sha256"]

            def send(request):
                nonlocal gateway, calls
                require(
                    request == job["request"] and input_pin is not None,
                    "C224_PRE_SEND_REQUEST_OR_INPUT_MISMATCH",
                )
                recorded = read_bound(input_pin)
                require(
                    recorded["runtime_identity"] == runtime
                    and recorded["question_sha256"] == job["question_sha256"]
                    and recorded["rendered_context_sha256"] == job["context_sha256"],
                    "C224_PERSISTED_INPUT_MISMATCH",
                )
                digest = job["request_payload_sha256"]
                started_path = receipts / (digest + ".started.json")
                response_path = receipts / (digest + ".response.json")
                body_path = receipts / (digest + ".body.bin")
                mode = "EXACT_REQUEST_RESPONSE_REUSE"

                if not response_path.exists():
                    require(
                        not started_path.exists() and not body_path.exists(),
                        "C224_PREVIOUS_ATTEMPT_NO_RESEND",
                    )
                    if gateway is None:
                        gateway = resolve_gateway()
                    endpoint, key, opener = gateway
                    with started_path.open("xb") as stream:
                        stream.write(encode({
                            "request_payload_sha256": digest,
                            "job_receipt": job_pin,
                            "input_receipt": input_pin,
                            "invocation_id": invocation,
                            "plan_receipt": plan_pin,
                            "attempt_number": 1,
                        }))
                        stream.flush()
                        os.fsync(stream.fileno())

                    outgoing = urllib.request.Request(
                        endpoint, data=encode(request), method="POST",
                        headers={
                            "Authorization": "apikey " + key,
                            "Content-Type": "application/json",
                        },
                    )
                    calls += 1
                    try:
                        with opener.open(outgoing, timeout=180) as response:
                            status, raw_body = response.status, response.read()
                    except urllib.error.HTTPError as error:
                        with error:
                            status, raw_body = error.code, error.read()
                    body_pin = write_once(body_path, raw_body)
                    write_once(response_path, encode({
                        "request_payload_sha256": digest,
                        "started_receipt_sha256": sha(started_path.read_bytes()),
                        "http_status": status,
                        "raw_body_receipt": body_pin,
                    }))
                    mode = "LIVE_DISPATCH_IN_THIS_PRODUCT_BOUNDARY"

                response_raw = response_path.read_bytes()
                response_record = json.loads(response_raw)
                started_raw = started_path.read_bytes()
                started_record = json.loads(started_raw)
                require(
                    response_record["request_payload_sha256"]
                    == started_record["request_payload_sha256"] == digest
                    and response_record["started_receipt_sha256"] == sha(started_raw)
                    and started_record["plan_receipt"] == plan_pin
                    and started_record["attempt_number"] == 1,
                    "C224_NETWORK_RECEIPT_CHAIN_MISMATCH",
                )
                read_bound(started_record["input_receipt"])
                raw_body = read_bytes(response_record["raw_body_receipt"])
                require(response_record["http_status"] == 200, "NON200_HTTP")
                envelope = json.loads(raw_body)
                choices = envelope.get("choices", [])
                require(len(choices) == 1, "EXPECTED_ONE_CHOICE")
                choice = choices[0]
                answer = choice.get("message", {}).get("content")
                require(
                    str(envelope.get("model", "")).strip().rsplit("/", 1)[-1].lower()
                    in {"deepseek-v4-flash", "deepseek-v4-flash-0731"},
                    "UNEXPECTED_RESPONSE_MODEL",
                )
                require(choice.get("finish_reason") != "length", "OUTPUT_TOKEN_LIMIT")
                require(
                    isinstance(answer, str) and bool(answer.strip()),
                    "EMPTY_OR_NON_TEXT_RESPONSE",
                )
                require(choice.get("finish_reason") == "stop", "INCOMPLETE_RESPONSE")
                state.update(
                    answer_sha=text_sha(answer),
                    response_mode=mode,
                    dispatch_invocation_id=started_record["invocation_id"],
                    response_receipt={
                        "path": response_path.relative_to(root).as_posix(),
                        "file_bytes_sha256": sha(response_raw),
                    },
                    usage_as_reported=envelope.get("usage"),
                    finish_reason=choice["finish_reason"],
                    response_model=envelope.get("model"),
                )
                events.append("complete_response_verified")
                return answer

            pipeline = object.__new__(product.RAGPipeline)
            pipeline.generator = candidate["ExplicitPolicyGenerator"](
                previous_request=job["parent_v2_request"],
                question=job["question"], context=job["context"],
                policy_path=root / policy_pin["path"], send=send,
            )
            returned = pipeline.ask_with_char_rrf(
                job["question"], candidate_id=job["candidate_id"],
                rank_sources=rank_sources, write_receipt=write_receipt, top_k=10,
            )
            require(
                output_pin is not None
                and returned["input_receipt_file_bytes_sha256"]
                == input_pin["file_bytes_sha256"]
                and returned["output_receipt_file_bytes_sha256"]
                == output_pin["file_bytes_sha256"],
                "C224_PRODUCT_RETURN_MISMATCH",
            )
            record = {
                "candidate_id": job["candidate_id"],
                "regression_spec_id": job["regression_spec_id"],
                "job_receipt": job_pin,
                "policy_receipt": policy_pin,
                "expected_behavior": job["expected_behavior"],
                "historical_disposition_preserved": job["historical_disposition"],
                "request_payload_sha256": job["request_payload_sha256"],
                "invocation_id": invocation,
                "input_receipt": input_pin,
                "output_receipt": output_pin,
                "answer": returned["answer"],
                "answer_text_sha256": state["answer_sha"],
                **state,
                "events": events,
                "execution_status": "COMPLETE_RESPONSE_PENDING_SCENARIO_REVIEW",
                "semantic_verdict": None,
                "product_constructor_executed": False,
            }
            write_once(checkpoint, encode(record))
            results.append(record)
            print(job["regression_spec_id"] + ": complete; review pending", flush=True)

        except Exception as error:
            blockers.append({
                "candidate_id": job["candidate_id"],
                "regression_spec_id": job["regression_spec_id"],
                "job_receipt": job_pin,
                "input_receipt": input_pin,
                "type": type(error).__name__,
                "detail": str(error),
                "status": "EXECUTION_GAP_NOT_POLICY_FAILURE",
                "automatic_resend": False,
            })
            print(job["regression_spec_id"] + ": execution gap; no resend", flush=True)

    require(calls <= unique_count, "C224_REQUEST_LIMIT_EXCEEDED")
    result = {
        "cell": "C224",
        "roadmap_subtask": "R11.3",
        "checklist_item": 12,
        "status": (
            "FIVE_SEEN_CLARIFICATION_RESPONSES_READY_FOR_REVIEW"
            if len(results) == 5 and not blockers else
            "SEEN_CLARIFICATION_EXECUTION_HAS_GAPS"
        ),
        "plan_receipt": plan_pin,
        "policy_receipt": policy_pin,
        "profile": profile,
        "scenario_count": 5,
        "completed_scenario_count": len(results),
        "candidate_results": results,
        "blocked_scenarios": blockers,
        "new_api_calls_attempted": calls,
        "historical_reference_estimate_toman": estimate,
        "current_rate_verified": False,
        "new_ranking_runs": 0,
        "new_model_pairs": 0,
        "scope": "SEEN_TEST_STORED_RANKS_OPT_IN_PRODUCT_BOUNDARY",
        "prompt_tuning_from_these_results_allowed": False,
        "independent_or_unseen_test_claimed": False,
        "active_default_route_changed": False,
        "historical_failures_overwritten": False,
        "new_semantic_passes": 0,
        "formal_nonregression_granted": False,
        "product_acceptance_granted": False,
        "done_item_count": 9,
        "open_item_count": 3,
        "open_item_ids": [9, 10, 12],
        "selected_version_gate": "BLOCKED",
    }
    result["output"] = save("C224.R11_seen_clarification_execution", result)
    print(json.dumps(result, ensure_ascii=False, indent=2))
    return result


V3_C224 = run_v3_c224()
````

### C225 — داوری پنج بازآزمون درخواست توضیح

در سه پاسخ، عبارت امتناع ثابت پیش از درخواست توضیح آمده است. این مورد با دستور نسخهٔ v2.1 برای حالت ابهام سازگار نیست و جدا از موفقیت اقدام روشن‌سازی ثبت می‌شود.

درخواست توضیح به‌صورت جملهٔ امری محترمانه نیز اقدام روشن‌سازی محسوب می‌شود؛ علامت سؤال معیار داوری نیست. معیارهای ازپیش‌ثبت‌شده تغییر نمی‌کنند.

توضیحات، مثال‌ها و استنادهای اضافی این پاسخ‌ها در این سل تأیید معنایی نمی‌شوند. نتایج نسخهٔ جدید با نتایج تاریخی یا پروفایل‌های دیگر ادغام نمی‌شوند.

هیچ prompt، پاسخ یا مرجع تغییر نمی‌کند. نتیجهٔ این Testهای دیده‌شده برای تنظیم دوبارهٔ prompt مصرف نمی‌شود و قبولی کلی محصول یا Gate صادر نمی‌شود.

````python
# NB22E-C226 — Four seen source-fidelity regressions, fixed candidate policy.
# C001 imports: Path, json, hashlib, os, tempfile, time, sys, importlib,
# urllib.parse/request/error, dotenv_values.

def run_v3_c226():
    def require(ok, code):
        if not ok:
            raise ValueError(code)

    def sha(raw):
        return hashlib.sha256(raw).hexdigest()

    def tsha(text):
        return sha(text.encode("utf-8"))

    def encode(value):
        return json.dumps(
            value, ensure_ascii=False, sort_keys=True,
            separators=(",", ":"), allow_nan=False,
        ).encode("utf-8")

    root = Path(
        "/Users/macbookpro/AMIR_DATA/00_WORKS/01_PROJECTS/"
        "RAG-projects/RAG_finance"
    ).resolve()
    folder = root / "reports/phase2/quality_improvement/NB22E.lab.v3"
    inputs = {}

    def read_bytes(pin):
        path = (root / pin["path"]).resolve()
        require(path.is_relative_to(root), "C226_PATH_OUTSIDE_PROJECT")
        raw = path.read_bytes()
        require(sha(raw) == pin["file_bytes_sha256"],
                "C226_HASH_MISMATCH:" + pin["path"])
        inputs[pin["path"]] = pin["file_bytes_sha256"]
        return raw

    def load(pin):
        return json.loads(read_bytes(pin))

    def report(stem, digest):
        return load({
            "path": folder.relative_to(root).as_posix()
                    + "/" + stem + "." + digest + ".json",
            "file_bytes_sha256": digest,
        })

    def write_once(path, raw):
        path.parent.mkdir(parents=True, exist_ok=True)
        if path.exists():
            require(path.read_bytes() == raw, "C226_OUTPUT_CONFLICT")
        else:
            fd, temporary = tempfile.mkstemp(prefix=".C226-", dir=str(path.parent))
            try:
                with os.fdopen(fd, "wb") as stream:
                    stream.write(raw)
                    stream.flush()
                    os.fsync(stream.fileno())
                try:
                    os.link(temporary, path)
                except FileExistsError:
                    require(path.read_bytes() == raw, "C226_OUTPUT_CONFLICT")
            finally:
                if os.path.exists(temporary):
                    os.unlink(temporary)
        require(path.read_bytes() == raw, "C226_RELOAD_CONFLICT")
        return {
            "path": path.relative_to(root).as_posix(),
            "file_bytes_sha256": sha(raw),
            "reload_validation": "PASS",
        }

    def save(stem, value):
        raw = encode(value)
        return write_once(folder / (stem + "." + sha(raw) + ".json"), raw)

    latest = report(
        "C225.R11_seen_clarification_review",
        "2815df52f1c2a07ccef80abd858fb07d7ee7286e23998797301c10e5271155c9",
    )
    ledger = report(
        "C198.R11_selected_citation_witness_disposition",
        "e9c605ecf404f084c0953b00f01e5d8dcb7752885aa720ffa83bf19c7789e6a3",
    )
    primary = report(
        "C191.R11_deepseek_batch_plan",
        "a1a590e38c04ac9f0a7431bd5aac9c972f77c97f86b0e3d318c81d17d9e87876",
    )
    packet = report(
        "C193.R11_batch_semantic_review_packet",
        "0b20b53e6812c69c47809a027f2c33c59e285bc6baf509db91d4e2707f460a4b",
    )
    setup = report(
        "C215.R11_versioned_answer_action_policy",
        "e5883cee3ffe3e62624a3e66e4a05b899c415f80ce74a83936aa5f931cc85756",
    )
    adapter_report = report(
        "C221.R11_candidate_policy_adapter_control",
        "0476d7db730dcd484c4b65a370c748cd5ea5480f207ad0e0a50ed3af7d4f0347",
    )
    policy_pin = latest["policy_receipt"]
    require(policy_pin == adapter_report["policy_receipt"],
            "C226_POLICY_VERSION_MISMATCH")
    load(policy_pin)

    def functions(pin):
        namespace = {"__name__": "c226_explicit_adapter"}
        exec(compile(read_bytes(pin).decode("utf-8"), pin["path"], "exec"),
             namespace)
        return namespace

    v2 = functions(setup["request_builder"])
    adapter = functions(adapter_report["module"])

    criteria = {
        "REG_0e7af55d68d4e6329a2b6958": {
            "historical_verdict": "SPECIFIC_SOURCE_SCOPE_GENERALIZED",
            "check": "PRESERVE_THE_SCOPE_AND_QUALIFIERS_OF_THE_CITED_SOURCE",
        },
        "REG_448f4e87d659a73b4079d755": {
            "historical_verdict": "LOAN_AMOUNT_AND_TERM_EXAMPLES_CONFLATED",
            "check": "DO_NOT_COMBINE_AMOUNT_AND_TERM_FROM_DIFFERENT_EXAMPLES",
        },
        "REG_36b7c2c4b802fdb7c498759a": {
            "historical_verdict": "PRODUCT_SPECIFIC_FACILITIES_SCOPE_OMITTED",
            "check": "KEEP_PRODUCT_IDENTITY_AND_APPLICABLE_FACILITY_CONDITIONS",
        },
        "REG_f2db9d585b476c6c8b82f83a": {
            "historical_verdict": "HOPEFUL_SOURCE_STATEMENT_STRENGTHENED",
            "check": "DO_NOT_STRENGTHEN_POSSIBILITY_OR_HOPE_INTO_CERTAINTY",
        },
    }
    old_rows = {
        row["regression_spec_id"]: row
        for row in ledger["scenario_dispositions"]
    }
    summaries = {row["candidate_id"]: row for row in primary["jobs"]}
    reviews = {row["candidate_id"]: row for row in packet["candidate_reviews"]}

    corpus_path = root / (
        "data/phase2/chunks/nb11_benchmark_v1/baseline_phase1_char_safe_v1.jsonl"
    )
    corpus_raw = corpus_path.read_bytes()
    require(
        sha(corpus_raw)
        == "3891c4ecad9332ba753cd605bbccd99da9d063f64934a30ee39e495349cffb50",
        "C226_CORPUS_CHANGED",
    )
    corpus = {
        row["chunk_id"]: row for row in (
            json.loads(line) for line in corpus_raw.splitlines() if line.strip()
        )
    }
    profile = {
        "model": "DeepSeek-V4-Flash", "temperature": 0,
        "max_tokens": 600, "thinking": {"type": "disabled"},
    }

    jobs = []
    for spec_id, criterion in criteria.items():
        old = old_rows[spec_id]
        cid = old["candidate_id"]
        require(
            old["scenario_disposition"] == "FAIL_SCOPED_EXPECTED_BEHAVIOR"
            and old["final_semantic_verdict"] == criterion["historical_verdict"],
            "C226_HISTORICAL_VERDICT_CHANGED",
        )
        previous_job = load(summaries[cid]["job_receipt"])
        prepared = load(previous_job["prepared_context_receipt"])
        rank = load(previous_job["rank_receipt"])
        historical_input = load(reviews[cid]["input_receipt"])
        require(
            spec_id in summaries[cid]["scenario_ids"]
            and tsha(prepared["question"]) == old["question_sha256"]
            and prepared["rendered_context"] == reviews[cid]["exact_dispatched_context"]
            and tsha(prepared["rendered_context"])
            == historical_input["rendered_context_sha256"]
            == old["rendered_context_sha256"]
            and historical_input["ordered_chunk_ids"] == rank["char_rrf_top10"],
            "C226_FIXED_INPUT_MISMATCH",
        )
        for witness in old.get("witnesses", []):
            require(
                prepared["rendered_context"][witness["start"]:witness["end"]]
                == witness["exact_quote"],
                "C226_BASELINE_WITNESS_MISMATCH",
            )

        parent_request = v2["build_request"](
            previous_job["request"], policy_path=root / setup["policy"]["path"]
        )
        request = adapter["build_request"](
            parent_request, policy_path=root / policy_pin["path"]
        )
        restored = json.loads(json.dumps(request))
        indices = [
            i for i, message in enumerate(restored["messages"])
            if message["role"] == "system"
        ]
        require(len(indices) == 1, "C226_SYSTEM_MESSAGE_NOT_UNIQUE")
        restored["messages"][indices[0]] = (
            previous_job["request"]["messages"][indices[0]]
        )
        require(
            restored == previous_job["request"]
            and {k: v for k, v in request.items() if k != "messages"} == profile,
            "C226_CHANGE_EXCEEDS_SYSTEM_PROMPT",
        )
        chunks = [
            {
                "chunk_id": chunk_id,
                "text": corpus[chunk_id]["content"],
                "title": corpus[chunk_id].get("title"),
                "url": corpus[chunk_id].get("url"),
            }
            for chunk_id in rank["char_rrf_top10"]
        ]
        require(
            [tsha(chunk["text"]) for chunk in chunks]
            == rank["char_rrf_top10_text_sha256"]
            == historical_input["ordered_chunk_text_sha256"],
            "C226_CHUNK_TEXT_MISMATCH",
        )
        job = {
            "candidate_id": cid,
            "regression_spec_id": spec_id,
            "question": prepared["question"],
            "question_sha256": old["question_sha256"],
            "context": prepared["rendered_context"],
            "context_sha256": old["rendered_context_sha256"],
            "chunks": chunks,
            "rank_receipt": previous_job["rank_receipt"],
            "parent_v2_request": parent_request,
            "request": request,
            "request_payload_sha256": sha(encode(request)),
            "expected_behavior": old["expected_behavior"],
            "historical_scenario": old,
            "review_focus": criterion["check"],
            "test_exposure": "PREVIOUSLY_SEEN_TEST",
        }
        jobs.append((job, save("C226.R11_frozen_fidelity_job", job)))

    require(len({job["request_payload_sha256"] for job, _ in jobs}) == 4,
            "C226_EXPECTED_FOUR_UNIQUE_REQUESTS")
    estimate = sum(
        ((len(encode(job["request"])) + 1024) * 43680 + 600 * 87360) / 1_000_000
        for job, _ in jobs
    )
    require(estimate < 100000, "C226_ESTIMATE_OUTSIDE_AUTO_BAND")
    print("C226 historical-reference estimate, toman:", round(estimate, 2))

    plan_pin = save("C226.R11_fidelity_run_plan", {
        "policy": policy_pin,
        "adapter": adapter_report["module"],
        "profile": profile,
        "jobs": [
            {
                "regression_spec_id": job["regression_spec_id"],
                "job_receipt": pin,
                "review_focus": job["review_focus"],
                "expected_behavior": job["expected_behavior"],
            }
            for job, pin in jobs
        ],
        "maximum_unique_attempts": 4,
        "historical_reference_estimate_toman": estimate,
        "current_rate_verified": False,
        "input_receipts": dict(inputs),
        "results_may_be_used_for_prompt_tuning": False,
    })
    receipts = folder / "C226.R11_receipts" / plan_pin["file_bytes_sha256"]
    receipts.mkdir(parents=True, exist_ok=True)

    product_path = root / "src/pipeline/rag_pipeline.py"
    require(
        sha(product_path.read_bytes())
        == "99ee048d00a7ad1884540fa297eb41df86ad4e54f6c49f26385ebb6b2ed504f6",
        "C226_PRODUCT_SOURCE_CHANGED",
    )
    if str(root) not in sys.path:
        sys.path.insert(0, str(root))
    product = importlib.import_module("src.pipeline.rag_pipeline")
    require(Path(product.__file__).resolve() == product_path.resolve(),
            "C226_PRODUCT_IMPORT_PATH_MISMATCH")

    gateway = None
    new_calls = 0

    def resolve_gateway():
        settings = dotenv_values(root / ".env", interpolate=False)
        base = str(os.environ.get("ARVAN_GATEWAY_URL")
                   or settings.get("ARVAN_GATEWAY_URL") or "").strip()
        key = str(os.environ.get("ARVAN_GATEWAY_API_KEY")
                  or settings.get("ARVAN_GATEWAY_API_KEY") or "").strip()
        url = urllib.parse.urlsplit(base)
        require(
            url.scheme == "https" and bool(url.hostname) and bool(key)
            and url.username is None and url.password is None
            and not url.query and not url.fragment,
            "C226_INVALID_GATEWAY",
        )
        route = url.path.rstrip("/") or "/v1"
        if not route.endswith("/chat/completions"):
            route += "/chat/completions"

        class NoRedirect(urllib.request.HTTPRedirectHandler):
            def redirect_request(self, req, fp, code, msg, headers, newurl):
                return None

        return (
            url._replace(path=route).geturl(), key,
            urllib.request.build_opener(NoRedirect()),
        )

    def execute_job(job, job_pin):
        nonlocal gateway, new_calls
        invocation = sha(
            (job_pin["file_bytes_sha256"] + ":" + str(time.time_ns())).encode()
        )
        input_pin = None
        output_pin = None
        state = {}
        runtime = {
            "scope": "SEEN_TEST_STORED_RANKS_FIXED_POLICY_FIDELITY",
            "invocation_id": invocation,
            "policy_sha256": policy_pin["file_bytes_sha256"],
            "rank_receipt_sha256": job["rank_receipt"]["file_bytes_sha256"],
            "job_sha256": job_pin["file_bytes_sha256"],
        }

        def rank_sources(*, question, candidate_id, top_k):
            require(
                candidate_id == job["candidate_id"] and top_k == 10
                and tsha(question) == job["question_sha256"],
                "C226_RUNTIME_QUERY_MISMATCH",
            )
            return {
                "candidate_id": candidate_id,
                "question_sha256": job["question_sha256"],
                "selected_variant": "char_rrf",
                "runtime_identity": runtime,
                "ordered_chunks": job["chunks"],
            }

        def write_receipt(role, payload):
            nonlocal input_pin, output_pin
            if role == "input":
                require(
                    input_pin is None
                    and payload["ordered_chunk_ids"]
                    == [chunk["chunk_id"] for chunk in job["chunks"]]
                    and payload["rendered_context_sha256"] == job["context_sha256"],
                    "C226_PRODUCT_INPUT_MISMATCH",
                )
                input_pin = save("C226.R11_product_input", payload)
                return input_pin["file_bytes_sha256"]
            require(
                role == "output" and input_pin is not None and output_pin is None
                and payload["input_receipt_file_bytes_sha256"]
                == input_pin["file_bytes_sha256"]
                and payload["answer_text_utf8_sha256"] == state["answer_sha"],
                "C226_PRODUCT_OUTPUT_MISMATCH",
            )
            output_pin = save("C226.R11_product_output", payload)
            return output_pin["file_bytes_sha256"]

        def send(request):
            nonlocal gateway, new_calls
            require(request == job["request"] and input_pin is not None,
                    "C226_PRE_SEND_MISMATCH")
            recorded = load(input_pin)
            require(
                recorded["runtime_identity"] == runtime
                and recorded["rendered_context_sha256"] == job["context_sha256"],
                "C226_PERSISTED_INPUT_MISMATCH",
            )
            digest = job["request_payload_sha256"]
            started_path = receipts / (digest + ".started.json")
            response_path = receipts / (digest + ".response.json")
            body_path = receipts / (digest + ".body.bin")
            mode = "EXACT_REQUEST_RESPONSE_REUSE"

            if not response_path.exists():
                require(
                    not started_path.exists() and not body_path.exists(),
                    "C226_PRIOR_ATTEMPT_NO_RESEND",
                )
                if gateway is None:
                    gateway = resolve_gateway()
                endpoint, key, opener = gateway
                with started_path.open("xb") as stream:
                    stream.write(encode({
                        "request_payload_sha256": digest,
                        "job_receipt": job_pin,
                        "input_receipt": input_pin,
                        "invocation_id": invocation,
                        "plan_receipt": plan_pin,
                        "attempt_number": 1,
                    }))
                    stream.flush()
                    os.fsync(stream.fileno())
                outgoing = urllib.request.Request(
                    endpoint, data=encode(request), method="POST",
                    headers={
                        "Authorization": "apikey " + key,
                        "Content-Type": "application/json",
                    },
                )
                new_calls += 1
                try:
                    with opener.open(outgoing, timeout=180) as response:
                        status, raw_body = response.status, response.read()
                except urllib.error.HTTPError as error:
                    with error:
                        status, raw_body = error.code, error.read()
                body_pin = write_once(body_path, raw_body)
                write_once(response_path, encode({
                    "request_payload_sha256": digest,
                    "started_receipt_sha256": sha(started_path.read_bytes()),
                    "http_status": status,
                    "raw_body_receipt": body_pin,
                }))
                mode = "LIVE_DISPATCH_IN_THIS_PRODUCT_BOUNDARY"

            response_raw = response_path.read_bytes()
            response_record = json.loads(response_raw)
            started_raw = started_path.read_bytes()
            started = json.loads(started_raw)
            require(
                response_record["request_payload_sha256"]
                == started["request_payload_sha256"] == digest
                and response_record["started_receipt_sha256"] == sha(started_raw)
                and started["plan_receipt"] == plan_pin
                and started["attempt_number"] == 1,
                "C226_TRANSPORT_CHAIN_MISMATCH",
            )
            require(response_record["http_status"] == 200, "NON200_HTTP")
            envelope = json.loads(read_bytes(response_record["raw_body_receipt"]))
            choices = envelope.get("choices", [])
            require(len(choices) == 1, "EXPECTED_ONE_CHOICE")
            choice = choices[0]
            answer = choice.get("message", {}).get("content")
            require(
                str(envelope.get("model", "")).strip().rsplit("/", 1)[-1].lower()
                in {"deepseek-v4-flash", "deepseek-v4-flash-0731"},
                "UNEXPECTED_RESPONSE_MODEL",
            )
            require(choice.get("finish_reason") != "length", "OUTPUT_TOKEN_LIMIT")
            require(isinstance(answer, str) and bool(answer.strip()),
                    "EMPTY_OR_NON_TEXT_RESPONSE")
            require(choice.get("finish_reason") == "stop", "INCOMPLETE_RESPONSE")
            state.update(
                answer_sha=tsha(answer),
                response_mode=mode,
                dispatch_invocation_id=started["invocation_id"],
                response_receipt={
                    "path": response_path.relative_to(root).as_posix(),
                    "file_bytes_sha256": sha(response_raw),
                },
                usage_as_reported=envelope.get("usage"),
                finish_reason=choice["finish_reason"],
            )
            return answer

        pipeline = object.__new__(product.RAGPipeline)
        pipeline.generator = adapter["ExplicitPolicyGenerator"](
            previous_request=job["parent_v2_request"],
            question=job["question"], context=job["context"],
            policy_path=root / policy_pin["path"], send=send,
        )
        returned = pipeline.ask_with_char_rrf(
            job["question"], candidate_id=job["candidate_id"],
            rank_sources=rank_sources, write_receipt=write_receipt, top_k=10,
        )
        require(
            output_pin is not None
            and returned["input_receipt_file_bytes_sha256"]
            == input_pin["file_bytes_sha256"]
            and returned["output_receipt_file_bytes_sha256"]
            == output_pin["file_bytes_sha256"],
            "C226_PRODUCT_RETURN_MISMATCH",
        )
        return {
            "candidate_id": job["candidate_id"],
            "regression_spec_id": job["regression_spec_id"],
            "job_receipt": job_pin,
            "policy_receipt": policy_pin,
            "request_payload_sha256": job["request_payload_sha256"],
            "invocation_id": invocation,
            "input_receipt": input_pin,
            "output_receipt": output_pin,
            "answer": returned["answer"],
            "answer_text_sha256": state["answer_sha"],
            **state,
            "execution_status": "COMPLETE_RESPONSE_PENDING_SCENARIO_REVIEW",
            "semantic_verdict": None,
            "product_constructor_executed": False,
        }

    results = []
    blockers = []
    for job, job_pin in jobs:
        checkpoint = receipts / (job_pin["file_bytes_sha256"] + ".completed.json")
        try:
            if checkpoint.exists():
                record = json.loads(checkpoint.read_bytes())
                require(
                    record["job_receipt"] == job_pin
                    and record["policy_receipt"] == policy_pin
                    and tsha(record["answer"]) == record["answer_text_sha256"],
                    "C226_COMPLETED_RECORD_MISMATCH",
                )
                recorded_input = load(record["input_receipt"])
                recorded_output = load(record["output_receipt"])
                require(
                    recorded_input["runtime_identity"]["invocation_id"]
                    == record["invocation_id"]
                    and recorded_output["input_receipt_file_bytes_sha256"]
                    == record["input_receipt"]["file_bytes_sha256"]
                    and recorded_output["answer_text_utf8_sha256"]
                    == record["answer_text_sha256"],
                    "C226_COMPLETED_CHAIN_MISMATCH",
                )
            else:
                record = execute_job(job, job_pin)
                write_once(checkpoint, encode(record))
            results.append(record)
            print(job["regression_spec_id"] + ": complete; review pending",
                  flush=True)
        except Exception as error:
            blockers.append({
                "regression_spec_id": job["regression_spec_id"],
                "job_receipt": job_pin,
                "type": type(error).__name__,
                "detail": str(error),
                "status": "EXECUTION_GAP_NOT_SEMANTIC_FAILURE",
                "automatic_resend": False,
            })
            print(job["regression_spec_id"] + ": execution gap; no resend",
                  flush=True)

    require(new_calls <= 4, "C226_ATTEMPT_LIMIT_EXCEEDED")
    result = {
        "cell": "C226",
        "roadmap_subtask": "R11.3",
        "checklist_item": 12,
        "status": (
            "FOUR_FIDELITY_RESPONSES_READY_FOR_REVIEW"
            if len(results) == 4 and not blockers
            else "FIDELITY_EXECUTION_HAS_GAPS"
        ),
        "plan_receipt": plan_pin,
        "policy_receipt": policy_pin,
        "profile": profile,
        "candidate_results": results,
        "blocked_scenarios": blockers,
        "completed_scenario_count": len(results),
        "new_api_calls_attempted": new_calls,
        "historical_reference_estimate_toman": estimate,
        "current_rate_verified": False,
        "new_ranking_runs": 0,
        "new_model_pairs": 0,
        "test_exposure": "PREVIOUSLY_SEEN_TEST",
        "prompt_changed": False,
        "prompt_tuning_from_results_allowed": False,
        "historical_results_overwritten": False,
        "active_default_route_changed": False,
        "new_semantic_passes": 0,
        "product_acceptance_granted": False,
        "formal_nonregression_granted": False,
        "done_item_count": 9,
        "open_item_count": 3,
        "open_item_ids": [9, 10, 12],
        "selected_version_gate": "BLOCKED",
    }
    result["output"] = save("C226.R11_seen_fidelity_execution", result)
    print(json.dumps(result, ensure_ascii=False, indent=2))
    return result


V3_C226 = run_v3_c226()
````

````python
# NB22E-C233 — Six seen-case, fixed-context policy executions.
# Uses C001 imports. No ranking, prompt tuning, or automatic retry.

def run_v3_c233():
    root = Path(
        "/Users/macbookpro/AMIR_DATA/00_WORKS/01_PROJECTS/"
        "RAG-projects/RAG_finance"
    ).resolve()
    folder = root / "reports/phase2/quality_improvement/NB22E.lab.v3"

    def require(ok, message):
        if not ok:
            raise ValueError(message)

    def sha(raw):
        return hashlib.sha256(raw).hexdigest()

    def encode(value):
        return (
            json.dumps(value, ensure_ascii=False, sort_keys=True, indent=2)
            + "\n"
        ).encode("utf-8")

    def read_pin(pin):
        path = root / pin["path"]
        require(path.is_file(), "C233_MISSING_FILE:" + str(path))
        raw = path.read_bytes()
        require(
            sha(raw) == pin["file_bytes_sha256"],
            "C233_HASH_MISMATCH:" + pin["path"],
        )
        return json.loads(raw.decode("utf-8"))

    def load(stem, digest):
        pin = {
            "path": str(
                (folder / f"{stem}.{digest}.json").relative_to(root)
            ),
            "file_bytes_sha256": digest,
        }
        return read_pin(pin), pin

    def save(stem, value):
        raw = encode(value)
        digest = sha(raw)
        path = folder / f"{stem}.{digest}.json"
        if path.exists():
            require(path.read_bytes() == raw, "C233_OUTPUT_COLLISION")
        else:
            with path.open("xb") as handle:
                handle.write(raw)
                handle.flush()
                os.fsync(handle.fileno())
        require(sha(path.read_bytes()) == digest, "C233_RELOAD_FAILED")
        return {
            "path": str(path.relative_to(root)),
            "file_bytes_sha256": digest,
        }

    c231, pin231 = load(
        "C231.R11_policy_precedence_disposition",
        "eb27f486b088510fbfc96968eecdbf02427fb1bd55685e7e2aed036fd0fc7ca7",
    )
    c232, pin232 = load(
        "C232.R11_case_policy_applicability",
        "ed006b8bbb11cbd086e44127fdf39d6855aec7bfe16e768a6973680310f62601",
    )
    packet, packet_pin = load(
        "C193.R11_batch_semantic_review_packet",
        "0b20b53e6812c69c47809a027f2c33c59e285bc6baf509db91d4e2707f460a4b",
    )
    batch, batch_pin = load(
        "C191.R11_deepseek_batch_plan",
        "a1a590e38c04ac9f0a7431bd5aac9c972f77c97f86b0e3d318c81d17d9e87876",
    )

    policy_pin = c232["input_receipts"]["policy"]
    policy = read_pin(policy_pin)
    prompt = policy["system_prompt"]
    require(
        sha(prompt.encode("utf-8")) == policy["system_prompt_sha256"],
        "C233_POLICY_PROMPT_HASH_MISMATCH",
    )

    expected = {}
    for row in c231["ledger"]:
        action = row.get("candidate_policy_expected_action")
        if action is not None:
            expected[row["regression_spec_id"]] = {
                "candidate_id": row["candidate_id"],
                "expected_action": action,
                "criteria": (
                    ["ASK_WHICH_FUND_TYPE_IS_INTENDED",
                     "DO_NOT_ASSERT_ONE_UNRESOLVED_INTERPRETATION",
                     "DO_NOT_USE_FIXED_ABSTENTION"]
                    if action == "ASK_RELEVANT_CLARIFICATION" else
                    ["ATTRIBUTE_THE_ANALYSIS_TO_THE_SOURCE",
                     "PRESERVE_ITS_STATED_LIMITATION",
                     "DO_NOT_PRESENT_THE_QUOTED_ANALYSIS_AS_UNIVERSAL_FACT"]
                ),
            }
    for row in c232["reviews"]:
        expected[row["regression_spec_id"]] = {
            "candidate_id": row["candidate_id"],
            "expected_action": row["policy_expected_action"],
            "criteria": row["confirmation_criteria"],
        }
    require(len(expected) == 6, "C233_EXPECTED_SIX_SCENARIOS")

    candidates = {
        row["candidate_id"]: row for row in packet["candidate_reviews"]
    }
    historical_jobs = {
        row["candidate_id"]: row for row in batch["jobs"]
    }
    jobs = {}

    for spec_id, criterion in sorted(expected.items()):
        candidate_id = criterion["candidate_id"]
        candidate = candidates[candidate_id]
        parent_pin = historical_jobs[candidate_id]["job_receipt"]
        parent = read_pin(parent_pin)
        request = json.loads(json.dumps(parent["request"]))

        require(
            request.get("model") == "DeepSeek-V4-Flash"
            and request.get("temperature") == 0
            and request.get("max_tokens") == 600
            and request.get("thinking") == {"type": "disabled"}
            and not request.get("stream", False),
            "C233_PARENT_PROFILE_MISMATCH",
        )
        systems = [
            message for message in request["messages"]
            if message["role"] == "system"
        ]
        require(len(systems) == 1, "C233_SYSTEM_MESSAGE_COUNT")
        require(
            sha(systems[0]["content"].encode("utf-8"))
            == "79694195488f4a8a31172f55cb37d28a84bf2b96c9a88c5ba426821b9419a41e",
            "C233_PARENT_PROMPT_MISMATCH",
        )

        user_text = "\n".join(
            message["content"] for message in request["messages"]
            if message["role"] == "user"
        )
        context = candidate["exact_dispatched_context"]
        require(
            context in user_text and candidate["question"] in user_text,
            "C233_FIXED_INPUT_NOT_FOUND_IN_PARENT_REQUEST",
        )

        systems[0]["content"] = prompt
        request_raw = encode(request)
        request_sha = sha(request_raw)

        if request_sha not in jobs:
            jobs[request_sha] = {
                "request": request,
                "request_payload_sha256": request_sha,
                "parent_job_receipt": parent_pin,
                "context_sha256": sha(context.encode("utf-8")),
                "scenarios": [],
            }
        jobs[request_sha]["scenarios"].append({
            "regression_spec_id": spec_id,
            **criterion,
        })

    require(len(jobs) <= 6, "C233_TOO_MANY_REQUESTS")

    plan = {
        "cell": "C233",
        "policy_receipt": policy_pin,
        "input_receipts": {
            "C231": pin231, "C232": pin232,
            "C193": packet_pin, "C191": batch_pin,
        },
        "scope": "SEEN_TEST_FIXED_CONTEXT_POLICY_EXECUTION",
        "request_serialization": "UTF8_SORTED_KEYS_INDENT2_TRAILING_NEWLINE",
        "jobs": list(jobs.values()),
        "maximum_attempts_per_request": 1,
        "automatic_retries": 0,
        "prompt_tuning_allowed": False,
    }
    plan_pin = save("C233.R11_six_case_policy_plan", plan)
    receipt_dir = folder / "C233.R11_receipts" / plan_pin["file_bytes_sha256"]
    receipt_dir.mkdir(parents=True, exist_ok=True)

    config = dict(dotenv_values(root / ".env", interpolate=False))
    config.update({
        key: value for key, value in os.environ.items()
        if value.strip()
    })
    base_url = str(config.get("ARVAN_GATEWAY_URL") or "").strip().rstrip("/")
    api_key = str(config.get("ARVAN_GATEWAY_API_KEY") or "").strip()
    require(base_url.startswith("https://"), "C233_HTTPS_GATEWAY_REQUIRED")
    require(bool(api_key), "C233_GATEWAY_KEY_MISSING")
    endpoint = (
        base_url if base_url.endswith("/chat/completions")
        else base_url + "/chat/completions"
    )

    results = []
    new_calls = 0

    for request_sha, job in jobs.items():
        started_path = receipt_dir / f"{request_sha}.started.json"
        completed_path = receipt_dir / f"{request_sha}.completed.json"
        raw_path = receipt_dir / f"{request_sha}.response.bin"

        if completed_path.exists():
            pin = json.loads(completed_path.read_text(encoding="utf-8"))
            record = read_pin(pin)
            require(
                record["request_payload_sha256"] == request_sha,
                "C233_REUSED_REQUEST_MISMATCH",
            )
            raw = (root / record["raw_response_receipt"]["path"]).read_bytes()
            require(
                sha(raw) == record["raw_response_receipt"]["file_bytes_sha256"],
                "C233_REUSED_RAW_RESPONSE_MISMATCH",
            )
            results.append(record)
            print(request_sha[:12] + ": saved result reused")
            continue

        if started_path.exists() or raw_path.exists():
            results.append({
                "request_payload_sha256": request_sha,
                "scenarios": job["scenarios"],
                "execution_status": "HELD_PREVIOUS_ATTEMPT_REQUIRES_RECOVERY",
                "behavior_verdict": None,
            })
            print(request_sha[:12] + ": held; previous attempt exists")
            continue

        started = {
            "plan_receipt": plan_pin,
            "request_payload_sha256": request_sha,
            "attempt_number": 1,
            "request": job["request"],
        }
        with started_path.open("xb") as handle:
            handle.write(encode(started))
            handle.flush()
            os.fsync(handle.fileno())

        body = encode(job["request"])
        request = urllib.request.Request(
            endpoint,
            data=body,
            headers={
                "Authorization": "Bearer " + api_key,
                "Content-Type": "application/json",
            },
            method="POST",
        )

        # Reject redirects so credentials are not forwarded elsewhere.
        class NoRedirect(urllib.request.HTTPRedirectHandler):
            def redirect_request(self, req, fp, code, msg, headers, newurl):
                return None

        opener = urllib.request.build_opener(NoRedirect())
        http_status = None
        transport_error = None
        new_calls += 1
        try:
            with opener.open(request, timeout=180) as response:
                http_status = response.status
                raw = response.read()
        except urllib.error.HTTPError as error:
            http_status = error.code
            raw = error.read()
        except Exception as error:
            raw = b""
            transport_error = type(error).__name__

        with raw_path.open("xb") as handle:
            handle.write(raw)
            handle.flush()
            os.fsync(handle.fileno())

        record = {
            "request_payload_sha256": request_sha,
            "scenarios": job["scenarios"],
            "plan_receipt": plan_pin,
            "context_sha256": job["context_sha256"],
            "http_status": http_status,
            "transport_error_type": transport_error,
            "raw_response_receipt": {
                "path": str(raw_path.relative_to(root)),
                "file_bytes_sha256": sha(raw),
            },
            "behavior_verdict": None,
            "execution_status": "EXECUTION_ERROR",
        }
        try:
            parsed = json.loads(raw.decode("utf-8"))
            choices = parsed.get("choices", [])
            choice = choices[0] if len(choices) == 1 else {}
            answer = choice.get("message", {}).get("content")
            model = str(parsed.get("model", "")).lower()
            finish = choice.get("finish_reason")
            record.update({
                "response_model": parsed.get("model"),
                "finish_reason": finish,
                "usage_as_reported": parsed.get("usage"),
                "answer": answer if isinstance(answer, str) else None,
            })
            if (
                http_status == 200
                and model in {"deepseek-v4-flash", "deepseek-v4-flash-0731"}
                and finish == "stop"
                and isinstance(answer, str)
                and bool(answer.strip())
            ):
                record["execution_status"] = "COMPLETE_RESPONSE_REVIEW_PENDING"
                record["answer_text_sha256"] = sha(answer.encode("utf-8"))
        except (ValueError, TypeError, UnicodeError):
            record["response_parse_error"] = True

        record_pin = save("C233.R11_six_case_response", record)
        with completed_path.open("x", encoding="utf-8") as handle:
            json.dump(record_pin, handle, ensure_ascii=False, indent=2)
            handle.flush()
            os.fsync(handle.fileno())

        results.append(record)
        print(request_sha[:12] + ": " + record["execution_status"])

    result = {
        "cell": "C233",
        "roadmap_subtask": "R11.3",
        "checklist_item": 12,
        "status": "SIX_CASE_POLICY_EXECUTION_RECORDED",
        "plan_receipt": plan_pin,
        "policy_receipt": policy_pin,
        "scope": "SEEN_TEST_FIXED_CONTEXT_POLICY_EXECUTION",
        "unique_request_count": len(jobs),
        "scenario_count": len(expected),
        "execution_counts": dict(Counter(
            item["execution_status"] for item in results
        )),
        "results": results,
        "new_api_calls_attempted": new_calls,
        "automatic_retries": 0,
        "new_ranking_runs": 0,
        "new_behavior_passes": 0,
        "product_runtime_execution_claimed": False,
        "historical_results_overwritten": False,
        "prompt_tuning_from_results_allowed": False,
        "active_default_route_changed": False,
        "formal_nonregression_granted": False,
        "done_item_count": 9,
        "open_item_count": 3,
        "open_item_ids": [9, 10, 12],
        "selected_version_gate": "BLOCKED",
    }
    output_pin = save("C233.R11_six_case_policy_execution", result)
    result["output"] = {**output_pin, "reload_validation": "PASS"}
    print(json.dumps(result, ensure_ascii=False, indent=2))
    return result


V3_C233 = run_v3_c233()
````

````python
# NB22E-C235 — Read existing R11.2 evidence from the exact notebook.
# No execution of notebook code; no repository scan or file mutation.

def run_v3_c235():
    root = Path(
        "/Users/macbookpro/AMIR_DATA/00_WORKS/01_PROJECTS/"
        "RAG-projects/RAG_finance"
    ).resolve()
    notebook_path = (
        root / "notebooks/phase2/22e_retrieval_experiment_lab_v3.ipynb"
    )

    if not notebook_path.is_file():
        raise FileNotFoundError(
            "C235_NOTEBOOK_MISSING:" + str(notebook_path)
        )

    raw = notebook_path.read_bytes()
    notebook = json.loads(raw.decode("utf-8"))

    def joined(value):
        return "".join(value) if isinstance(value, list) else str(value or "")

    def mentions_metrics_subtask(text):
        return bool(re.search(
            r"(?<![A-Za-z0-9])R11(?:\.|_|٫)2(?![0-9])",
            text,
            flags=re.IGNORECASE,
        ))

    evidence = []
    for index, cell in enumerate(notebook.get("cells", [])):
        source = joined(cell.get("source", ""))
        outputs = []

        for output in cell.get("outputs", []):
            if output.get("output_type") == "error":
                continue

            if "text" in output:
                outputs.append(joined(output["text"]))

            data = output.get("data", {})
            if "application/json" in data:
                outputs.append(json.dumps(
                    data["application/json"], ensure_ascii=False, indent=2
                ))
            elif "text/plain" in data:
                outputs.append(joined(data["text/plain"]))

        relevant_outputs = [
            text for text in outputs if mentions_metrics_subtask(text)
        ]
        relevant_markdown = (
            cell.get("cell_type") == "markdown"
            and mentions_metrics_subtask(source)
        )

        # Code is never executed or treated as proof of a successful run.
        if relevant_outputs or relevant_markdown:
            evidence.append({
                "notebook_cell_index": index,
                "cell_type": cell.get("cell_type"),
                "execution_count": cell.get("execution_count"),
                "markdown": source if relevant_markdown else None,
                "matching_recorded_outputs": relevant_outputs,
            })

    result = {
        "cell": "C235",
        "status": (
            "EXISTING_R11_2_RECORDS_READY_FOR_REVIEW"
            if evidence else
            "NO_EXPLICIT_R11_2_RECORD_FOUND_IN_SAVED_NOTEBOOK"
        ),
        "notebook_receipt": {
            "path": str(notebook_path.relative_to(root)),
            "file_bytes_sha256": hashlib.sha256(raw).hexdigest(),
        },
        "scope": "EXPLICIT_R11_2_MENTIONS_IN_ONE_SAVED_NOTEBOOK",
        "absence_is_repository_wide": False,
        "unsaved_kernel_state_examined": False,
        "matching_cell_count": len(evidence),
        "evidence": evidence,
        "new_api_calls": 0,
        "new_model_runs": 0,
        "new_ranking_runs": 0,
        "new_metric_runs": 0,
        "files_written": 0,
        "R11_3_complete": False,
        "selected_version_gate": "BLOCKED",
    }

    print(json.dumps(result, ensure_ascii=False, indent=2))
    return result


V3_C235 = run_v3_c235()
````

````python
# NB22E-C236 — Read exact metric evidence and contract dependencies.
# No model/API/metric execution and no repository scan.

def run_v3_c236():
    root = Path(
        "/Users/macbookpro/AMIR_DATA/00_WORKS/01_PROJECTS/"
        "RAG-projects/RAG_finance"
    ).resolve()

    pins = {
        "NB22E_C025_historical_evidence_binding": {
            "path": (
                "reports/phase2/quality_improvement/NB22E.lab.v3/"
                "C025.R11_recovered_evidence_binding."
                "2fdd2e89e91c54c59076fca8b63760a3bf8e0c6546e13740a11733b9b670f55f.json"
            ),
            "file_bytes_sha256":
                "2fdd2e89e91c54c59076fca8b63760a3bf8e0c6546e13740a11733b9b670f55f",
        },
        "NB22E_C033_gate_disposition": {
            "path": (
                "reports/phase2/quality_improvement/NB22E.lab.v3/"
                "C033.R11_gate_disposition."
                "e62892a30721c5668efaaf23b5decdf8bd683cc05fcbabc463a40fe1f23a5b53.json"
            ),
            "file_bytes_sha256":
                "e62892a30721c5668efaaf23b5decdf8bd683cc05fcbabc463a40fe1f23a5b53",
        },
        "NB21C_C025_execution_contract": {
            "path": (
                "reports/phase2/evaluation/NB21C.C025.action12_contract/"
                "C025.execution_contract."
                "edc54bb349c632e3424aceb8da881d124d681786bf0f384be0dffdfca0ae2a6b.json"
            ),
            "file_bytes_sha256":
                "edc54bb349c632e3424aceb8da881d124d681786bf0f384be0dffdfca0ae2a6b",
        },
    }

    documents = {}
    missing = []
    for name, pin in pins.items():
        path = root / pin["path"]
        if not path.is_file():
            missing.append({
                "path": str(path),
                "dependent_decision": name,
            })
            continue

        raw = path.read_bytes()
        if hashlib.sha256(raw).hexdigest() != pin["file_bytes_sha256"]:
            raise ValueError("C236_HASH_MISMATCH:" + pin["path"])
        documents[name] = json.loads(raw.decode("utf-8"))

    # Preserve all dictionary fields and ordinary lists.
    # Avoid reprinting hundreds of historical case bindings.
    def readable(value):
        if isinstance(value, dict):
            return {key: readable(item) for key, item in value.items()}
        if isinstance(value, list):
            if len(value) > 40:
                fields = sorted({
                    key
                    for item in value if isinstance(item, dict)
                    for key in item
                })
                return {
                    "display_only_summary": True,
                    "item_count": len(value),
                    "item_dictionary_fields": fields,
                    "items_not_printed": True,
                }
            return [readable(item) for item in value]
        return value

    # Expose explicit file references without resolving unrelated files.
    references = []

    def collect_references(value, source_name, location="$"):
        if isinstance(value, dict):
            if isinstance(value.get("path"), str):
                references.append({
                    "source": source_name,
                    "json_location": location,
                    "path": value["path"],
                    "file_bytes_sha256": value.get("file_bytes_sha256"),
                })
            for key, item in value.items():
                collect_references(item, source_name, location + "." + key)
        elif isinstance(value, list):
            # Large case lists are not a new discovery/search scope.
            if len(value) <= 40:
                for index, item in enumerate(value):
                    collect_references(
                        item, source_name, f"{location}[{index}]"
                    )

    for name, document in documents.items():
        collect_references(document, name)

    result = {
        "cell": "C236",
        "roadmap_subtask": "R11.2",
        "status": (
            "BOUND_METRIC_DOCUMENTS_READY_FOR_READINESS_REVIEW"
            if not missing else
            "EXACT_REQUIRED_DOCUMENT_MISSING"
        ),
        "verified_receipts": {
            name: pins[name] for name in documents
        },
        "missing_exact_files": missing,
        "documents": {
            name: readable(document)
            for name, document in documents.items()
        },
        "explicit_document_references": references,
        "scope": "THREE_EXPLICIT_RECEIPTS_ONLY",
        "historical_bindings_promoted_to_selected_metrics": False,
        "new_api_calls": 0,
        "new_model_runs": 0,
        "new_metric_runs": 0,
        "files_written": 0,
        "R11_3_complete": False,
        "selected_version_gate": "BLOCKED",
    }

    print(json.dumps(result, ensure_ascii=False, indent=2))
    return result


V3_C236 = run_v3_c236()
````

````python
# NB22E-C237 — Existing-attempt accounting plus metric-gap disposition.
# Uses C001 imports. No network, generation, or ranking.

def run_v3_c237():
    root = Path(
        "/Users/macbookpro/AMIR_DATA/00_WORKS/01_PROJECTS/"
        "RAG-projects/RAG_finance"
    ).resolve()
    folder = root / "reports/phase2/quality_improvement/NB22E.lab.v3"

    def require(ok, message):
        if not ok:
            raise ValueError(message)

    def sha(raw):
        return hashlib.sha256(raw).hexdigest()

    def load(stem, digest, base=None):
        path = (base or folder) / f"{stem}.{digest}.json"
        pin = {
            "path": str(path.relative_to(root)),
            "file_bytes_sha256": digest,
        }
        return read_json(pin), pin

    def read_bytes(pin):
        path = root / pin["path"]
        require(path.is_file(), "C237_MISSING_FILE:" + str(path))
        raw = path.read_bytes()
        require(
            sha(raw) == pin["file_bytes_sha256"],
            "C237_HASH_MISMATCH:" + pin["path"],
        )
        return raw

    def read_json(pin):
        return json.loads(read_bytes(pin).decode("utf-8"))

    execution, execution_pin = load(
        "C233.R11_six_case_policy_execution",
        "ff9988dcf4d3a002dc003d081491bf6bd395f998e47e431fdf43a80c3db54750",
    )
    binding, binding_pin = load(
        "C025.R11_recovered_evidence_binding",
        "2fdd2e89e91c54c59076fca8b63760a3bf8e0c6546e13740a11733b9b670f55f",
    )
    contract, contract_pin = load(
        "C025.execution_contract",
        "edc54bb349c632e3424aceb8da881d124d681786bf0f384be0dffdfca0ae2a6b",
        root / "reports/phase2/evaluation/NB21C.C025.action12_contract",
    )
    plan = read_json(execution["plan_receipt"])
    jobs = {
        job["request_payload_sha256"]: job for job in plan["jobs"]
    }
    require(len(jobs) == 6, "C237_EXPECTED_SIX_UNIQUE_REQUESTS")

    attempts = []
    seen = set()
    fixed_abstention = "در منابع موجود پاسخ کافی پیدا نشد."

    for record in execution["results"]:
        request_sha = record["request_payload_sha256"]
        require(
            request_sha in jobs and request_sha not in seen,
            "C237_UNKNOWN_OR_DUPLICATE_ATTEMPT",
        )
        seen.add(request_sha)
        job = jobs[request_sha]

        request_raw = (
            json.dumps(
                job["request"], ensure_ascii=False,
                sort_keys=True, indent=2,
            ) + "\n"
        ).encode("utf-8")
        require(sha(request_raw) == request_sha, "C237_REQUEST_HASH_MISMATCH")

        raw_pin = record["raw_response_receipt"]
        raw_path = root / raw_pin["path"]
        require(
            raw_path.name == request_sha + ".response.bin",
            "C237_RESPONSE_FILENAME_MISMATCH",
        )
        started_path = raw_path.with_name(request_sha + ".started.json")
        require(started_path.is_file(), "C237_STARTED_RECEIPT_MISSING")
        started_raw = started_path.read_bytes()
        started = json.loads(started_raw.decode("utf-8"))
        require(
            started["request_payload_sha256"] == request_sha
            and started["attempt_number"] == 1
            and started["request"] == job["request"]
            and started["plan_receipt"] == execution["plan_receipt"],
            "C237_ATTEMPT_BINDING_MISMATCH",
        )

        response = read_json(raw_pin)
        require(
            record["scenarios"] == job["scenarios"],
            "C237_SCENARIO_BINDING_MISMATCH",
        )

        choices = response.get("choices", [])
        choice = choices[0] if len(choices) == 1 else {}
        answer = choice.get("message", {}).get("content")
        completed = (
            record.get("http_status") == 200
            and choice.get("finish_reason") == "stop"
            and isinstance(answer, str)
            and bool(answer.strip())
            and str(response.get("model", "")).lower()
            in {"deepseek-v4-flash", "deepseek-v4-flash-0731"}
        )
        require(
            completed
            and answer == record["answer"]
            and sha(answer.encode("utf-8"))
            == record["answer_text_sha256"],
            "C237_COMPLETE_RESPONSE_BINDING_MISMATCH",
        )

        usage = response.get("usage")
        require(
            isinstance(usage, dict)
            and usage == record["usage_as_reported"],
            "C237_USAGE_BINDING_MISMATCH",
        )

        token_values = {}
        for field in ("prompt_tokens", "completion_tokens", "total_tokens"):
            value = usage.get(field)
            require(
                type(value) is int and value >= 0,
                "C237_INVALID_USAGE_FIELD:" + field,
            )
            token_values[field] = value

        require(
            token_values["total_tokens"]
            == token_values["prompt_tokens"] + token_values["completion_tokens"],
            "C237_TOKEN_TOTAL_MISMATCH",
        )

        attempts.append({
            "request_payload_sha256": request_sha,
            "scenario_ids": [
                item["regression_spec_id"] for item in job["scenarios"]
            ],
            "started_receipt": {
                "path": str(started_path.relative_to(root)),
                "file_bytes_sha256": sha(started_raw),
            },
            "raw_response_receipt": raw_pin,
            "generation_outcome": "COMPLETE_TEXT_RESPONSE",
            "usage_as_reported": usage,
            "accounted_tokens": token_values,
            "reported_cost": usage.get("cost"),
            "reported_cost_currency": None,
            "cost_interpretation": "NOT_ASSUMED_WITHOUT_DOCUMENTED_COST_BASIS",
            "latency_seconds": None,
            "latency_status": "NO_BOUND_ELAPSED_DURATION_USED",
            "contains_abstention_text": fixed_abstention in answer,
            "abstention_presence_is_execution_failure": False,
        })

    require(seen == set(jobs), "C237_ATTEMPT_COVERAGE_MISMATCH")
    totals = {
        field: sum(row["accounted_tokens"][field] for row in attempts)
        for field in ("prompt_tokens", "completion_tokens", "total_tokens")
    }

    gap_rows = binding["existing_reconciliation"]["metric_reconciliation"]
    metric_policy = contract["metric_policy"]
    metric_ledger = []

    for old in gap_rows:
        name = old["metric"]
        require(name in metric_policy, "C237_METRIC_NOT_IN_CONTRACT:" + name)
        rule = metric_policy[name]

        local_evidence = None
        if name == "token_usage":
            local_evidence = {
                "scope": "C233_SIX_UNIQUE_PROVIDER_ATTEMPTS_ONLY",
                "attempt_count": len(attempts),
                "usage_covered_attempt_count": len(attempts),
                "totals": totals,
                "cached_or_reasoning_details_added_again": False,
            }
        elif name == "failure_rate":
            local_evidence = {
                "scope": "C233_FIXED_CONTEXT_GENERATION_ATTEMPTS_ONLY",
                "attempted_count": len(attempts),
                "complete_text_count": len(attempts),
                "generation_error_count": 0,
                "outcome_unknown_count": 0,
                "attempt_generation_error_rate": 0.0,
                "end_to_end_product_failure_rate": None,
                "semantic_failures_counted_as_generation_errors": False,
            }

        metric_ledger.append({
            "metric": name,
            "contract_rule": rule,
            "historical_status": old["historical_status"],
            "existing_requirement_fa": old["remaining_requirement_fa"],
            "existing_evidence": old["existing_evidence"],
            "new_scoped_accounting": local_evidence,
            "selected_population_metric_status":
                "NOT_ESTABLISHED_BY_THIS_SCOPED_ACCOUNTING",
            "formal_nonregression_status": "NOT_COMPARABLE_NOT_PASS",
            "gap_closed": False,
        })

    result = {
        "cell": "C237",
        "roadmap_subtask": "R11.2",
        "status": "EXISTING_ATTEMPT_ACCOUNTING_COMPLETE_FORMAL_METRIC_GAPS_EXPLICIT",
        "input_receipts": {
            "C233": execution_pin,
            "historical_metric_binding": binding_pin,
            "NB21C_C025_contract": contract_pin,
        },
        "scope": "SIX_SEEN_TEST_FIXED_CONTEXT_REQUESTS_ONLY",
        "unique_attempt_count": len(attempts),
        "scenario_count": sum(len(row["scenario_ids"]) for row in attempts),
        "complete_text_response_count": len(attempts),
        "generation_error_count": 0,
        "outcome_unknown_count": 0,
        "provider_reported_token_totals": totals,
        "token_usage_coverage": "6/6",
        "interpretable_cost_total": None,
        "cost_unknown_is_zero": False,
        "latency_quantiles": None,
        "attempts": attempts,
        "metric_gap_ledger": metric_ledger,
        "formal_metric_gaps_closed": 0,
        "historical_results_overwritten": False,
        "semantic_verdicts_changed": False,
        "new_api_calls": 0,
        "new_model_runs": 0,
        "new_ranking_runs": 0,
        "new_semantic_judgments": 0,
        "new_scoped_accounting_metrics": ["token_usage", "failure_rate"],
        "formal_metric_runs": 0,
        "R11_2_complete": False,
        "R11_3_complete": False,
        "selected_version_gate": "BLOCKED",
    }

    raw = (
        json.dumps(result, ensure_ascii=False, indent=2) + "\n"
    ).encode("utf-8")
    digest = sha(raw)
    path = folder / f"C237.R11_scoped_operational_accounting.{digest}.json"
    if path.exists():
        require(path.read_bytes() == raw, "C237_EXISTING_OUTPUT_MISMATCH")
    else:
        with path.open("xb") as handle:
            handle.write(raw)
            handle.flush()
            os.fsync(handle.fileno())

    require(sha(path.read_bytes()) == digest, "C237_RELOAD_FAILED")
    result["output"] = {
        "path": str(path.relative_to(root)),
        "file_bytes_sha256": digest,
        "reload_validation": "PASS",
    }
    print(json.dumps(
        {
            key: value for key, value in result.items()
            if key not in {"attempts", "metric_gap_ledger"}
        },
        ensure_ascii=False, indent=2,
    ))
    return result


V3_C237 = run_v3_c237()
````

````python
# NB22E-C240 — Approved planning package; no network/ranking/bootstrap.
# Uses C001 imports.

def run_v3_c240():
    root = Path(
        "/Users/macbookpro/AMIR_DATA/00_WORKS/01_PROJECTS/"
        "RAG-projects/RAG_finance"
    ).resolve()
    folder = root / "reports/phase2/quality_improvement/NB22E.lab.v3"

    def require(ok, code):
        if not ok:
            raise ValueError(code)

    def sha(raw):
        return hashlib.sha256(raw).hexdigest()

    def enc(value):
        return (
            json.dumps(value, ensure_ascii=False, sort_keys=True, indent=2)
            + "\n"
        ).encode("utf-8")

    receipts = {}

    def read_pin(pin, role):
        path = (root / pin["path"]).resolve()
        require(path.is_relative_to(root), "C240_PATH_OUTSIDE_PROJECT")
        require(path.is_file(), "C240_MISSING_FILE:" + str(path))
        raw = path.read_bytes()
        require(sha(raw) == pin["file_bytes_sha256"],
                "C240_HASH_MISMATCH:" + pin["path"])
        receipts[role] = pin
        return json.loads(raw.decode("utf-8"))

    def lab(stem, digest, role):
        return read_pin({
            "path": str((folder / f"{stem}.{digest}.json").relative_to(root)),
            "file_bytes_sha256": digest,
        }, role)

    manifest = lab(
        "C239.planning_input_manifest",
        "a8d3c1ab4fb3143524fcfec2e58a9a2147ec9ccc120100d51f7d6a587892498c",
        "C239_manifest",
    )
    initial = {}
    for pin in manifest["included_receipts"]:
        name = Path(pin["path"]).name
        if name.endswith(".json"):
            initial[name.split(".")[0] + ":" + name.split(".")[1]] = read_pin(
                pin, name
            )

    report = initial["C070:report"]
    results209 = initial["C209:R11_fixed_pool_local_completion"]
    pool_plan = initial["C207:R11_fixed_pool_arm_plan"]
    contract = initial["C025:execution_contract"]

    components = read_pin(
        report["outputs"]["source_components"], "C070_source_components"
    )
    paired = read_pin(report["outputs"]["paired_cases"], "C070_paired_cases")

    statistical_pins = [
        pin for pin in report["input_bindings"].values()
        if ".statistical_plan." in pin["path"]
    ]
    require(len(statistical_pins) == 1, "C240_STATISTICAL_PLAN_NOT_UNIQUE")
    statistical = read_pin(statistical_pins[0], "C063_statistical_plan")

    # Reconstruct the complete provenance graph, never from score values.
    case_by_id = {row["candidate_id"]: row for row in paired}
    require(len(case_by_id) == len(paired) == 1149,
            "C240_TEST_POPULATION_MISMATCH")

    parent = {candidate_id: candidate_id for candidate_id in case_by_id}

    def find(candidate_id):
        while parent[candidate_id] != candidate_id:
            parent[candidate_id] = parent[parent[candidate_id]]
            candidate_id = parent[candidate_id]
        return candidate_id

    source_owner = {}
    for row in paired:
        source_ids = row["reference_source_ids"]
        require(
            isinstance(source_ids, list) and bool(source_ids)
            and all(isinstance(item, str) and item for item in source_ids),
            "C240_UNKNOWN_REFERENCE_SOURCE",
        )
        for source_id in source_ids:
            if source_id in source_owner:
                parent[find(row["candidate_id"])] = find(source_owner[source_id])
            else:
                source_owner[source_id] = row["candidate_id"]

    reconstructed = {}
    for candidate_id in case_by_id:
        reconstructed.setdefault(find(candidate_id), set()).add(candidate_id)

    stored_sets = [frozenset(members) for members in components.values()]
    require(
        len(stored_sets) == len(set(stored_sets))
        and sum(len(members) for members in stored_sets) == 1149
        and set(stored_sets)
        == {frozenset(members) for members in reconstructed.values()},
        "C240_SOURCE_COMPONENT_PARTITION_MISMATCH",
    )

    membership = {}
    for group_id, members in components.items():
        for candidate_id in members:
            require(candidate_id not in membership,
                    "C240_DUPLICATE_GROUP_MEMBERSHIP")
            membership[candidate_id] = group_id
            require(
                case_by_id[candidate_id]["source_component"] == group_id,
                "C240_PAIRED_CASE_GROUP_MISMATCH",
            )

    population = pool_plan["population"]
    target_ids = sorted({row["candidate_id"] for row in population})
    require(len(population) == 104 and len(target_ids) == 70,
            "C240_FIXED_POOL_POPULATION_MISMATCH")
    require(set(target_ids) <= set(membership),
            "C240_TARGET_GROUP_COVERAGE_INCOMPLETE")

    scenario_ids = {row["regression_spec_id"] for row in population}
    require(
        scenario_ids == {
            row["regression_spec_id"] for row in results209["scenario_results"]
        },
        "C240_C209_SCENARIO_IDENTITY_MISMATCH",
    )

    grouping_rows = [{
        "candidate_id": candidate_id,
        "source_component": membership[candidate_id],
        "reference_source_ids":
            case_by_id[candidate_id]["reference_source_ids"],
        "full_component_size": len(components[membership[candidate_id]]),
        "membership_receipt": receipts["C070_source_components"],
        "reference_lineage_receipt": receipts["C070_paired_cases"],
        "producer": "NB21C.C070_ACTION40",
        "component_method_receipt": receipts["C063_statistical_plan"],
    } for candidate_id in target_ids]

    group_counts = Counter(row["source_component"] for row in grouping_rows)
    require(len(group_counts) == 58, "C240_EXPECTED_GROUP_COUNT_CHANGED")

    # Bind the prospective response-comparison population.
    batch = lab(
        "C191.R11_deepseek_batch_plan",
        "a1a590e38c04ac9f0a7431bd5aac9c972f77c97f86b0e3d318c81d17d9e87876",
        "C191",
    )
    packet = lab(
        "C193.R11_batch_semantic_review_packet",
        "0b20b53e6812c69c47809a027f2c33c59e285bc6baf509db91d4e2707f460a4b",
        "C193",
    )
    policy_pin = {
        "path": "src/generation/policies/answer_action_policy_v2_1_candidate.json",
        "file_bytes_sha256":
            "e3a664516ce943d6254d655987962d16c1142cd72f58be5c2fae1a0a29d6948b",
    }
    policy = read_pin(policy_pin, "candidate_policy")
    require(
        sha(policy["system_prompt"].encode("utf-8"))
        == policy["system_prompt_sha256"],
        "C240_CANDIDATE_PROMPT_HASH_MISMATCH",
    )

    candidates = {
        row["candidate_id"]: row for row in packet["candidate_reviews"]
    }
    reference_requests, candidate_requests = {}, {}
    request_population = []

    for summary in batch["jobs"]:
        candidate_id = summary["candidate_id"]
        job = read_pin(summary["job_receipt"], "parent_job:" + candidate_id)
        old = json.loads(json.dumps(job["request"]))
        systems = [m for m in old["messages"] if m["role"] == "system"]
        require(len(systems) == 1, "C240_SYSTEM_COUNT")
        require(
            sha(systems[0]["content"].encode("utf-8"))
            == "79694195488f4a8a31172f55cb37d28a84bf2b96c9a88c5ba426821b9419a41e",
            "C240_REFERENCE_PROMPT_MISMATCH",
        )
        require(
            old.get("model") == "DeepSeek-V4-Flash"
            and old.get("max_tokens") == 600
            and old.get("temperature") == 0
            and old.get("thinking") == {"type": "disabled"},
            "C240_PROFILE_MISMATCH",
        )

        new = json.loads(json.dumps(old))
        [m for m in new["messages"] if m["role"] == "system"][0]["content"] = (
            policy["system_prompt"]
        )
        old_id, new_id = sha(enc(old)), sha(enc(new))
        reference_requests[old_id] = old
        candidate_requests[new_id] = new
        request_population.append({
            "candidate_id": candidate_id,
            "scenario_ids": summary["scenario_ids"],
            "question_sha256": candidates[candidate_id]["question_sha256"],
            "context_sha256": candidates[candidate_id]["rendered_context_sha256"],
            "reference_logical_request_sha256": old_id,
            "candidate_logical_request_sha256": new_id,
            "parent_job_receipt": summary["job_receipt"],
        })

    require(
        len(request_population) == 33
        and len(reference_requests) == len(candidate_requests) == 32
        and len({
            spec for row in request_population for spec in row["scenario_ids"]
        }) == 35,
        "C240_RESPONSE_POPULATION_CHANGED",
    )

    # Inventory exact-request evidence; this is not formal reuse acceptance.
    observed = {"reference": set(), "candidate": set()}
    reuse_evidence = []

    def response_json(pin):
        value = read_pin(pin, "response:" + pin["file_bytes_sha256"])
        if "choices" in value:
            return value
        if isinstance(value.get("raw_body_receipt"), dict):
            raw_pin = value["raw_body_receipt"]
            path = root / raw_pin["path"]
            raw = path.read_bytes()
            require(sha(raw) == raw_pin["file_bytes_sha256"],
                    "C240_RAW_BODY_HASH_MISMATCH")
            return json.loads(raw.decode("utf-8"))
        if isinstance(value.get("response_body"), str):
            return json.loads(value["response_body"])
        return None

    def record_reuse(arm, request, answer, response_pin, source):
        response = response_json(response_pin)
        if response is None:
            reuse_evidence.append({
                "source": source, "response_receipt": response_pin,
                "status": "RESPONSE_ENVELOPE_REQUIRES_EXPLICIT_ADAPTER",
            })
            return

        choices = response.get("choices", [])
        complete = (
            len(choices) == 1
            and choices[0].get("finish_reason") == "stop"
            and isinstance(answer, str) and bool(answer.strip())
            and choices[0].get("message", {}).get("content") == answer
            and str(response.get("model", "")).lower()
            in {"deepseek-v4-flash", "deepseek-v4-flash-0731"}
        )
        identity = sha(enc(request))
        allowed = reference_requests if arm == "reference" else candidate_requests
        if complete and identity in allowed:
            observed[arm].add(identity)
            reuse_evidence.append({
                "arm": arm,
                "logical_request_sha256": identity,
                "answer_text_sha256": sha(answer.encode("utf-8")),
                "response_receipt": response_pin,
                "source": source,
                "status": "EXACT_PAYLOAD_AND_COMPLETE_RESPONSE_MATCH",
                "formal_provenance_equivalence_accepted": False,
            })

    for row in request_population:
        candidate_id = row["candidate_id"]
        historical = candidates[candidate_id]
        if historical.get("execution_status") == "COMPLETE":
            record_reuse(
                "reference",
                reference_requests[row["reference_logical_request_sha256"]],
                historical["answer"], historical["response_receipt"], "C192/C193",
            )

    for stem, digest in [
        (
            "C224.R11_seen_clarification_execution",
            "fd254f8e4822bd9541fdc4c26009a965d2b4f164d6d963528908feb6a4ce9220",
        ),
        (
            "C226.R11_seen_fidelity_execution",
            "2eabe09e4ef0abf7a918434f4d0c5782c3f362b8ca9f4880c72d4dc498718207",
        ),
    ]:
        execution = lab(stem, digest, stem)
        for record in execution["candidate_results"]:
            job = read_pin(
                record["job_receipt"],
                "candidate_job:" + record["candidate_id"],
            )
            record_reuse(
                "candidate", job["request"], record["answer"],
                record["response_receipt"], stem,
            )

    latest = lab(
        "C233.R11_six_case_policy_execution",
        "ff9988dcf4d3a002dc003d081491bf6bd395f998e47e431fdf43a80c3db54750",
        "C233",
    )
    latest_plan = read_pin(latest["plan_receipt"], "C233_plan")
    latest_jobs = {
        row["request_payload_sha256"]: row for row in latest_plan["jobs"]
    }
    for record in latest["results"]:
        record_reuse(
            "candidate",
            latest_jobs[record["request_payload_sha256"]]["request"],
            record["answer"], record["raw_response_receipt"], "C233",
        )

    missing = {
        "reference": sorted(set(reference_requests) - observed["reference"]),
        "candidate": sorted(set(candidate_requests) - observed["candidate"]),
    }

    # Historical planning estimate, not a verified current quote or hard cap.
    requests_by_arm = {
        "reference": reference_requests, "candidate": candidate_requests,
    }
    estimated_input_tokens = {}
    for arm, requests in requests_by_arm.items():
        for identity, request in requests.items():
            # UTF-8 byte count is a disclosed planning proxy, not tokenization.
            estimated_input_tokens[(arm, identity)] = sum(
                len(str(message["content"]).encode("utf-8"))
                for message in request["messages"]
            )

    input_rate = batch["input_toman_per_million"]
    output_rate = batch["output_toman_per_million"]

    def estimate(jobs):
        return {
            "request_count": len(jobs),
            "maximum_requested_output_tokens": 600 * len(jobs),
            "estimated_input_tokens_proxy": sum(
                estimated_input_tokens[item] for item in jobs
            ),
            "historical_estimate_toman": (
                sum(estimated_input_tokens[item] for item in jobs) * input_rate
                + 600 * len(jobs) * output_rate
            ) / 1_000_000,
        }

    all_jobs = [
        (arm, identity)
        for arm, requests in requests_by_arm.items()
        for identity in requests
    ]
    unmatched_jobs = [
        (arm, identity) for arm, identities in missing.items()
        for identity in identities
    ]

    decisions = [
        {
            "id": "RESPONSE_COMPARISON_SCOPE",
            "decision_needed": (
                "قبول یا اصلاح دامنهٔ مقایسهٔ prompt نسخهٔ اول و "
                "نسخهٔ نامزد ۲٫۱ روی بستهٔ ۳۵ سناریویی دیده‌شده؛ "
                "این دامنه پذیرش کل نسخهٔ گسترش‌یافته نیست."
            ),
        },
        {
            "id": "COMMON_QUALIFIED_RUBRIC",
            "decision_needed": (
                "مرجع و Rubric مشترکِ واجد صلاحیت، واحدهای ادعا، "
                "نیازمندی، استناد و Safety پیش از اجرای وابسته "
                "تثبیت شوند؛ انتظار هر بازو جدا و پسینی تغییر نکند."
            ),
        },
        {
            "id": "REUSE_PROVENANCE",
            "decision_needed": (
                "هم‌ارزی منشأ اجرای درخواست‌های منطبق احراز شود؛ "
                "انطباق payload به‌تنهایی هم‌ارزی Runtime یا "
                "خط پایهٔ تاریخی پذیرفته‌شده نیست."
            ),
        },
        {
            "id": "TEST_GROUPING_METHOD",
            "decision_needed": (
                "مصرف نگاشت اصلی C070 برای زیرجمعیت با حفظ "
                "وابستگی‌های کل گراف و اتصال نسخهٔ سؤال/مرجع کنترل "
                "شود؛ نگاشت جدیدی جایگزین نشده است."
            ),
        },
        {
            "id": "STATISTICAL_CONFIGURATION",
            "decision_needed": (
                "تفاوت seed قرارداد اصلی و طرح اقدام ۴۰ صریح "
                "تعیین‌تکلیف شود؛ عدد یا روش مطلوب از نتیجه انتخاب نشود."
            ),
            "original_contract": contract["uncertainty_policy"],
            "action40_plan": statistical["uncertainty"],
        },
        {
            "id": "EXECUTION_READINESS",
            "decision_needed": (
                "پس از تعیین شروط بالا، تعداد درخواست کم‌شده، "
                "پروفایل، منشأ تعرفه و مجوز اجرا نهایی شوند. "
                "خطاهای قبلی پاک یا خودکار دوباره ارسال نشوند."
            ),
        },
    ]

    package = {
        "cell": "C240",
        "status": "PROSPECTIVE_PLAN_AND_EXISTING_GROUPING_READY_FOR_DECISION",
        "grouping": {
            "producer": "NB21C.C070_ACTION40",
            "method_producer": "NB21C.C063_STATISTICAL_PLAN",
            "full_test_query_count": 1149,
            "full_component_count": len(components),
            "full_partition_reconstructed_from_source_ids": True,
            "full_partition_matches_stored_components": True,
            "target_query_count": 70,
            "covered_target_query_count": len(grouping_rows),
            "target_component_count": len(group_counts),
            "target_component_size_distribution":
                dict(Counter(group_counts.values())),
            "membership_rows": grouping_rows,
            "full_components_preserved": True,
            "unknown_group_assumed_independent": False,
            "new_grouping_created": False,
            "grouping_unit_changed": False,
            "bootstrap_executed": False,
            "formal_method_acceptance_granted": False,
        },
        "prospective_response_comparison": {
            "component": "FIXED_CONTEXT_GENERATION_SYSTEM_PROMPT",
            "arms": {
                "reference": {
                    "arm_id": "prompt_v1_reference_comparator_v1",
                    "historically_accepted_baseline": False,
                    "system_prompt_sha256":
                        "79694195488f4a8a31172f55cb37d28a84bf2b96c9a88c5ba426821b9419a41e",
                },
                "candidate": {
                    "arm_id": "policy_v2_1_fixed_context_candidate_v1",
                    "policy_receipt": policy_pin,
                },
            },
            "difference": "SYSTEM_PROMPT_ONLY",
            "fixed_conditions": [
                "QUESTION", "ORDERED_CONTEXT_AND_SOURCE_TEXT",
                "USER_WRAPPER", "MODEL", "TEMPERATURE",
                "OUTPUT_CAP", "THINKING_SETTING",
                "COMMON_REFERENCE_AND_SCORING_RULE",
                "DECLARED_GENERATION_BOUNDARY",
            ],
            "profile": {
                "model": "DeepSeek-V4-Flash", "temperature": 0,
                "max_tokens": 600, "thinking": {"type": "disabled"},
            },
            "population": request_population,
            "candidate_count": 33,
            "scenario_count": 35,
            "logical_requests_per_arm": 32,
            "maximum_requests_if_no_prior_result_qualifies": 64,
            "exposure": "PREVIOUSLY_SEEN_TEST",
            "selection_rationale": "ENTIRE_PREEXISTING_C191_BATCH_NOT_OUTCOME_FILTERED",
            "savings_case": "OUTSIDE_THIS_C191_BATCH_NOT_SILENTLY_MERGED",
            "independent_holdout_claimed": False,
            "prompt_tuning_allowed": False,
            "request_identity_serialization":
                "SORTED_UTF8_JSON_INDENT2_TRAILING_NEWLINE",
            "existing_exact_complete_request_counts": {
                arm: len(values) for arm, values in observed.items()
            },
            "reuse_evidence": reuse_evidence,
            "requests_without_verified_complete_match": missing,
            "reuse_is_conditional_on_provenance_acceptance": True,
            "metric_rules": contract["metric_policy"],
            "coverage_rules": contract["coverage_policy"],
            "comparison_rules": contract["regression_policy"],
            "uncertainty_rules": contract["uncertainty_policy"],
            "metric_denominators": "CONTRACT_UNITS_NOT_NUMBER_OF_SCENARIOS",
            "different_profiles_or_versions_pooled": False,
            "semantic_reference_qualification": "NOT_ESTABLISHED_BY_THIS_PLAN",
            "formal_acceptance": "BLOCKED_UNTIL_COMMON_REFERENCE_AND_DESIGN_QUALIFY",
        },
        "fixed_pool_inference_plan": {
            "component": "POOL50_RANK_ORDER_ONLY",
            "reference_arm": "SAVED_RERANKER_POOL50_ORDER_VERSION_BOUND_IN_C207",
            "candidate_arm": pool_plan["configuration"],
            "population": population,
            "unique_query_count": 70,
            "scenario_count": 104,
            "existing_execution_receipt":
                next(pin for pin in receipts.values()
                     if "C209.R11_fixed_pool_local_completion." in pin["path"]),
            "ranking_jobs_required": 0,
            "generation_jobs_required": 0,
            "api_calls_required": 0,
            "metrics_scope": [
                "hit_at_10", "recall_at_10", "ndcg_at_10"
            ],
            "full_population_absolute_acceptance_implied": False,
            "retrospective_miss_selected_subset_generalization": False,
            "historical_31_met_73_not_met_preserved": True,
            "bootstrap_pending_method_and_configuration_decision": True,
            "transfer_to_expanded_or_156_group": False,
        },
        "cost_plan": {
            "tariff_source": receipts["C191"],
            "rate_basis": "C042_HISTORICAL_REFERENCE_RATES_AS_RECORDED_IN_C191",
            "input_toman_per_million": input_rate,
            "output_toman_per_million": output_rate,
            "input_estimation_method": "MESSAGE_CONTENT_UTF8_BYTE_COUNT_PROXY",
            "proxy_is_tokenizer_measurement_or_guaranteed_bound": False,
            "all_requests_scenario": estimate(all_jobs),
            "conditional_unmatched_requests_scenario": estimate(unmatched_jobs),
            "current_tariff_verified": False,
            "invoice_or_charge_claimed": False,
            "local_work_time_cost": "NOT_MEASURED",
        },
        "remaining_execution_decisions": decisions,
        "input_receipts": receipts,
        "package_prepared_for_decision": True,
        "execution_ready": False,
        "new_api_calls": 0,
        "new_model_runs": 0,
        "new_ranking_runs": 0,
        "bootstrap_runs": 0,
        "gold_changes": 0,
        "historical_results_overwritten": False,
        "R11_3_open_item_ids": [9, 10, 12],
        "selected_version_gate": "BLOCKED",
    }

    raw = enc(package)
    digest = sha(raw)
    path = folder / f"C240.R11_prospective_comparison_and_test_groups.{digest}.json"
    if path.exists():
        require(path.read_bytes() == raw, "C240_OUTPUT_COLLISION")
    else:
        with path.open("xb") as handle:
            handle.write(raw)
            handle.flush()
            os.fsync(handle.fileno())
    require(sha(path.read_bytes()) == digest, "C240_RELOAD_FAILED")

    summary = {
        "cell": "C240",
        "status": package["status"],
        "group_coverage": "70/70",
        "target_component_count": len(group_counts),
        "full_graph_partition_verified": True,
        "new_grouping_created": False,
        "response_comparison_scenarios": 35,
        "unique_requests_per_arm": 32,
        "existing_exact_complete_request_counts":
            package["prospective_response_comparison"][
                "existing_exact_complete_request_counts"
            ],
        "conditional_missing_request_count": len(unmatched_jobs),
        "cost_plan": package["cost_plan"],
        "remaining_execution_decisions": decisions,
        "execution_ready": False,
        "new_api_calls": 0,
        "new_ranking_runs": 0,
        "bootstrap_runs": 0,
        "selected_version_gate": "BLOCKED",
        "output": {
            "path": str(path.relative_to(root)),
            "file_bytes_sha256": digest,
            "reload_validation": "PASS",
        },
    }
    print(json.dumps(summary, ensure_ascii=False, indent=2))
    return package


V3_C240 = run_v3_c240()
````

````python
# NB22E-C241 — Mechanical deficit audit and common-rubric proposal.
# No network, generation, ranking, Bootstrap, or prompt change.

def run_v3_c241():
    root = Path(
        "/Users/macbookpro/AMIR_DATA/00_WORKS/01_PROJECTS/"
        "RAG-projects/RAG_finance"
    ).resolve()
    folder = root / "reports/phase2/quality_improvement/NB22E.lab.v3"

    def require(ok, message):
        if not ok:
            raise ValueError(message)

    def sha(raw):
        return hashlib.sha256(raw).hexdigest()

    def enc(value):
        return (
            json.dumps(value, ensure_ascii=False, sort_keys=True, indent=2)
            + "\n"
        ).encode("utf-8")

    verified = {}

    def raw_pin(pin):
        path = (root / pin["path"]).resolve()
        require(path.is_relative_to(root), "C241_PATH_OUTSIDE_PROJECT")
        require(path.is_file(), "C241_MISSING_FILE:" + str(path))
        raw = path.read_bytes()
        require(sha(raw) == pin["file_bytes_sha256"],
                "C241_HASH_MISMATCH:" + pin["path"])
        verified[pin["path"]] = pin
        return raw

    def read_pin(pin):
        return json.loads(raw_pin(pin).decode("utf-8"))

    def load(stem, digest):
        pin = {
            "path": str((folder / f"{stem}.{digest}.json").relative_to(root)),
            "file_bytes_sha256": digest,
        }
        return read_pin(pin), pin

    def save(stem, value):
        raw = enc(value)
        digest = sha(raw)
        path = folder / f"{stem}.{digest}.json"
        if path.exists():
            require(path.read_bytes() == raw, "C241_OUTPUT_COLLISION")
        else:
            with path.open("xb") as handle:
                handle.write(raw)
                handle.flush()
                os.fsync(handle.fileno())
        require(sha(path.read_bytes()) == digest, "C241_RELOAD_FAILED")
        return {
            "path": str(path.relative_to(root)),
            "file_bytes_sha256": digest,
            "reload_validation": "PASS",
        }

    package, package_pin = load(
        "C240.R11_prospective_comparison_and_test_groups",
        "3f68803a338bc43597de67732e6895f96267eebe7b4942f23885a4059022a108",
    )
    packet, packet_pin = load(
        "C193.R11_batch_semantic_review_packet",
        "0b20b53e6812c69c47809a027f2c33c59e285bc6baf509db91d4e2707f460a4b",
    )
    history, history_pin = load(
        "C198.R11_selected_citation_witness_disposition",
        "e9c605ecf404f084c0953b00f01e5d8dcb7752885aa720ffa83bf19c7789e6a3",
    )
    applicability, applicability_pin = load(
        "C232.R11_case_policy_applicability",
        "ed006b8bbb11cbd086e44127fdf39d6855aec7bfe16e768a6973680310f62601",
    )
    precedence, precedence_pin = load(
        "C231.R11_policy_precedence_disposition",
        "eb27f486b088510fbfc96968eecdbf02427fb1bd55685e7e2aed036fd0fc7ca7",
    )
    batch, batch_pin = load(
        "C191.R11_deepseek_batch_plan",
        "a1a590e38c04ac9f0a7431bd5aac9c972f77c97f86b0e3d318c81d17d9e87876",
    )

    comparison = package["prospective_response_comparison"]
    policy_pin = comparison["arms"]["candidate"]["policy_receipt"]
    policy = read_pin(policy_pin)
    require(
        sha(policy["system_prompt"].encode("utf-8"))
        == policy["system_prompt_sha256"],
        "C241_POLICY_PROMPT_MISMATCH",
    )

    candidates = {r["candidate_id"]: r for r in packet["candidate_reviews"]}
    historical = {
        r["regression_spec_id"]: r for r in history["scenario_dispositions"]
    }
    requests = {}
    candidates_by_request = {}
    for row in comparison["population"]:
        parent = read_pin(row["parent_job_receipt"])
        reference = parent["request"]
        candidate = json.loads(json.dumps(reference))
        systems = [m for m in candidate["messages"] if m["role"] == "system"]
        require(len(systems) == 1, "C241_SYSTEM_COUNT")
        systems[0]["content"] = policy["system_prompt"]

        for arm, request in [("reference", reference), ("candidate", candidate)]:
            identity = sha(enc(request))
            require(
                identity == row[arm + "_logical_request_sha256"],
                "C241_REQUEST_IDENTITY_CHANGED",
            )
            key = (arm, identity)
            requests[key] = request
            candidates_by_request.setdefault(key, set()).add(row["candidate_id"])

    require(len(requests) == 64, "C241_EXPECTED_64_ARM_REQUESTS")

    # Read-only adapter: unwrap explicit containers; do not alter response fields.
    def unwrap(pin):
        value = read_pin(pin)
        steps = []
        for _ in range(5):
            if isinstance(value, dict) and "choices" in value:
                return value, steps, None
            if not isinstance(value, dict):
                return None, steps, "NON_OBJECT_RESPONSE_ENVELOPE"

            if isinstance(value.get("raw_body_receipt"), dict):
                value = json.loads(
                    raw_pin(value["raw_body_receipt"]).decode("utf-8")
                )
                steps.append("READ_HASH_BOUND_RAW_BODY")
                continue

            selected = []
            for key in ("response_body", "body", "response", "raw_response"):
                item = value.get(key)
                if isinstance(item, dict) and "choices" in item:
                    selected.append((key, item))
                elif isinstance(item, str):
                    try:
                        parsed = json.loads(item)
                    except (ValueError, TypeError):
                        continue
                    if isinstance(parsed, dict) and "choices" in parsed:
                        selected.append((key, parsed))

            if len(selected) != 1:
                return None, steps, "NO_UNIQUE_EXPLICIT_RESPONSE_CONTAINER"
            key, value = selected[0]
            steps.append("UNWRAP_" + key)
        return None, steps, "ENVELOPE_DEPTH_LIMIT"

    observations = {key: [] for key in requests}

    def observe(arm, request, answer, pin, source):
        key = (arm, sha(enc(request)))
        if key not in observations:
            return
        try:
            response, steps, unresolved = unwrap(pin)
        except (ValueError, UnicodeError, FileNotFoundError) as error:
            observations[key].append({
                "source": source, "receipt": pin,
                "state": "UNRESOLVED",
                "reason": type(error).__name__ + ":" + str(error),
            })
            return

        if unresolved:
            observations[key].append({
                "source": source, "receipt": pin,
                "state": "UNRESOLVED", "reason": unresolved,
                "adapter_steps": steps,
            })
            return

        choices = response.get("choices", [])
        choice = choices[0] if len(choices) == 1 else {}
        text = choice.get("message", {}).get("content")
        model = str(response.get("model", "")).lower()
        complete = (
            len(choices) == 1
            and choice.get("finish_reason") == "stop"
            and isinstance(text, str) and bool(text.strip())
            and model in {"deepseek-v4-flash", "deepseek-v4-flash-0731"}
        )
        if complete and isinstance(answer, str) and text != answer:
            state, reason = "UNRESOLVED", "RECORDED_AND_RAW_ANSWER_MISMATCH"
        elif complete:
            state, reason = "COMPLETE", None
        elif choice.get("finish_reason") == "length" or not text:
            state, reason = "INCOMPLETE", "NO_COMPLETE_TEXT_RESPONSE"
        else:
            state, reason = "UNRESOLVED", "MODEL_OR_FINISH_NOT_VERIFIED"

        observations[key].append({
            "source": source,
            "receipt": pin,
            "state": state,
            "reason": reason,
            "adapter_steps": steps,
            "finish_reason": choice.get("finish_reason"),
            "response_model": response.get("model"),
            "answer_text_sha256": (
                sha(text.encode("utf-8")) if isinstance(text, str) else None
            ),
            "usage_as_reported": response.get("usage"),
        })

    # Reference: include incomplete records too; C240 had skipped them.
    for row in comparison["population"]:
        historical_response = candidates[row["candidate_id"]]
        observe(
            "reference",
            requests[("reference", row["reference_logical_request_sha256"])],
            historical_response.get("answer"),
            historical_response["response_receipt"],
            "C192/C193",
        )

    for stem, digest in [
        (
            "C224.R11_seen_clarification_execution",
            "fd254f8e4822bd9541fdc4c26009a965d2b4f164d6d963528908feb6a4ce9220",
        ),
        (
            "C226.R11_seen_fidelity_execution",
            "2eabe09e4ef0abf7a918434f4d0c5782c3f362b8ca9f4880c72d4dc498718207",
        ),
    ]:
        document, _ = load(stem, digest)
        for record in document["candidate_results"]:
            job = read_pin(record["job_receipt"])
            observe(
                "candidate", job["request"], record.get("answer"),
                record["response_receipt"], stem,
            )

    latest, _ = load(
        "C233.R11_six_case_policy_execution",
        "ff9988dcf4d3a002dc003d081491bf6bd395f998e47e431fdf43a80c3db54750",
    )
    latest_plan = read_pin(latest["plan_receipt"])
    latest_jobs = {
        r["request_payload_sha256"]: r for r in latest_plan["jobs"]
    }
    for record in latest["results"]:
        observe(
            "candidate",
            latest_jobs[record["request_payload_sha256"]]["request"],
            record.get("answer"),
            record["raw_response_receipt"],
            "C233",
        )

    deficits = {
        (arm, identity)
        for arm, identities in comparison[
            "requests_without_verified_complete_match"
        ].items()
        for identity in identities
    }
    require(len(deficits) == 21, "C241_EXPECTED_21_C240_DEFICITS")

    def classify(key):
        items = observations[key]
        complete = [r for r in items if r["state"] == "COMPLETE"]
        if complete:
            states = {r["answer_text_sha256"] for r in complete}
            if len(states) > 1:
                return "ORIGIN_OR_FORMAT_UNRESOLVED", "MULTIPLE_DISTINCT_COMPLETE_ANSWERS"
            if any(not r["adapter_steps"] for r in complete):
                return "REUSABLE", "EXACT_REQUEST_COMPLETE_RESPONSE"
            return "MECHANICAL_ADAPTER_REQUIRED", "READ_ONLY_UNWRAPPING_VERIFIED"
        if any(r["state"] == "UNRESOLVED" for r in items):
            return "ORIGIN_OR_FORMAT_UNRESOLVED", "EXPLICIT_EVIDENCE_REQUIRES_RESOLUTION"
        return (
            "FRESH_EXECUTION_REQUIRED",
            "ONLY_INCOMPLETE_RESPONSES" if items
            else "NO_RESULT_IN_ENUMERATED_BOUND_PRODUCER_RECORDS",
        )

    audit = [{
        "arm": key[0],
        "logical_request_sha256": key[1],
        "candidate_ids": sorted(candidates_by_request[key]),
        "category": classify(key)[0],
        "reason": classify(key)[1],
        "observations": observations[key],
        "formal_runtime_equivalence_accepted": False,
    } for key in sorted(deficits)]

    # Common rubric proposal; never derived from the new answers above.
    overrides = {}
    for row in precedence["ledger"]:
        action = row.get("candidate_policy_expected_action")
        if action:
            overrides[row["regression_spec_id"]] = action
    for row in applicability["reviews"]:
        overrides[row["regression_spec_id"]] = row["policy_expected_action"]

    action_checks = {
        "ASK_RELEVANT_CLARIFICATION": [
            "ASK_A_SPECIFIC_QUESTION_RESOLVING_THE_MATERIAL_AMBIGUITY",
            "DO_NOT_ASSERT_AN_UNRESOLVED_INTERPRETATION_AS_CERTAIN",
            "DO_NOT_USE_FIXED_ABSTENTION_FOR_MATERIAL_AMBIGUITY",
        ],
        "ANSWER_SUPPORTED": [
            "ANSWER_THE_QUESTION_DIRECTLY_FROM_THE_FIXED_CONTEXT",
            "PRESERVE_SOURCE_SCOPE_AND_QUALIFIERS",
            "DO_NOT_OMIT_REQUIRED_SUPPORTED_INFORMATION",
        ],
        "BOUNDED_ANSWER_SUPPORTED": [
            "STATE_THE_SUPPORTED_INTERPRETATION_OR_SCOPE_EXPLICITLY",
            "ANSWER_WITHIN_THAT_SCOPE",
            "PRESERVE_SOURCE_LIMITATIONS_AND_UNCERTAINTY",
        ],
    }
    rubric = []
    for candidate in packet["candidate_reviews"]:
        context = candidate["exact_dispatched_context"]
        for spec in candidate["scenario_specs"]:
            spec_id = spec["regression_spec_id"]
            old = historical[spec_id]
            witnesses = old.get("witnesses", [])
            verified_witnesses = []
            issues = []

            for witness in witnesses:
                if (
                    isinstance(witness.get("start"), int)
                    and isinstance(witness.get("end"), int)
                    and context[witness["start"]:witness["end"]]
                    == witness["exact_quote"]
                ):
                    verified_witnesses.append(witness)
                else:
                    issues.append("PRIOR_WITNESS_NOT_BOUND_TO_THIS_CONTEXT")

            original_action = spec.get(
                "behavior_expectation", {}
            ).get("expected_behavior")
            action = overrides.get(spec_id, original_action)

            if action in {
                "ANSWER_WITH_EXPLICIT_SOURCE_SCOPE",
                "ANSWER_WITH_SOURCE_ATTRIBUTION_AND_LIMITATION",
            }:
                action = "BOUNDED_ANSWER_SUPPORTED"
            if action == "CLARIFICATION_REQUIRED":
                action = "ASK_RELEVANT_CLARIFICATION"

            if old["scenario_disposition"] == "BLOCKED_REFERENCE_SCOPE_NOT_ESTABLISHED":
                issues.append("QUESTION_REFERENCE_SCOPE_NOT_ESTABLISHED")

            if spec["kind"] == "CITATION_WITNESS":
                action = "ASSESS_CURRENT_CLAIM_LINK_CONDITIONALLY"
                checks = [
                    "FIRST_IDENTIFY_WHETHER_THE_HISTORICAL_CLAIM_LINK_IS_RETAINED",
                    "IF_RETAINED_BIND_CURRENT_CLAIM_TO_QUALIFIED_SOURCE_AND_EXACT_SPAN",
                    "IF_NOT_RETAINED_RECORD_NOT_APPLICABLE_NOT_A_CITATION_PASS",
                    "PRESERVE_UNSUPPORTED_OR_UNDETERMINED_WITH_REASON",
                ]
                evidence = spec["historical_link_expectation"]
                issues.append("CURRENT_LINK_APPLICABILITY_IS_RESPONSE_DEPENDENT")
            else:
                checks = action_checks.get(action)
                evidence = {
                    "context_sha256": sha(context.encode("utf-8")),
                    "prior_exact_witnesses": verified_witnesses,
                    "proposed_bounded_content": old.get("bounded_answer_fa"),
                    "prior_judgments_are_auxiliary_only": True,
                }
                if checks is None:
                    checks = []
                    issues.append("EXPECTED_ACTION_NOT_DEFINED")
                if (
                    action in {"ANSWER_SUPPORTED", "BOUNDED_ANSWER_SUPPORTED"}
                    and not verified_witnesses
                ):
                    issues.append("CASE_SPECIFIC_CONTENT_WITNESS_NOT_BOUND")

            blocking = [
                issue for issue in issues
                if issue != "CURRENT_LINK_APPLICABILITY_IS_RESPONSE_DEPENDENT"
            ]
            rubric.append({
                "regression_spec_id": spec_id,
                "candidate_id": candidate["candidate_id"],
                "question": candidate["question"],
                "original_specification": spec,
                "expected_action": action,
                "required_evidence": evidence,
                "common_acceptance_criteria": checks,
                "unresolved_issues": issues,
                "rubric_status": (
                    "BLOCKED_EXPECTATION_OR_EVIDENCE"
                    if blocking else "PROPOSED_FOR_REFERENCE_APPROVAL"
                ),
                "same_criteria_for_both_arms": True,
                "historical_expectation_preserved": True,
                "official_reference_approved": False,
            })

    require(
        len(rubric) == 35
        and len({r["regression_spec_id"] for r in rubric}) == 35,
        "C241_RUBRIC_POPULATION_MISMATCH",
    )

    rubric_document = {
        "schema_version": "NB22E_COMMON_POLICY_RUBRIC_PROPOSAL_V1",
        "scope": "C191_35_SEEN_SCENARIOS_ONLY",
        "basis": {
            "policy": policy_pin,
            "original_specs_in_C193": packet_pin,
            "prior_auxiliary_judgments": history_pin,
            "applicability": applicability_pin,
            "precedence": precedence_pin,
        },
        "rules": {
            "same_criteria_for_both_arms": True,
            "post_dispatch_criterion_changes_allowed": False,
            "current_answers_used_to_choose_criteria": False,
            "prior_exposure_disclosed": True,
            "independent_reference_claimed": False,
            "execution_error_is_policy_failure": False,
            "unassessable_or_not_applicable_is_pass": False,
            "full_response_acceptance_requires_assessment_of_additional_claims": True,
        },
        "rows": rubric,
        "approval_status": "PROPOSAL_NOT_OFFICIAL_REFERENCE",
    }
    rubric_pin = save("C241.R11_common_policy_rubric_proposal", rubric_document)

    fresh_keys = {
        key for key in deficits
        if classify(key)[0] == "FRESH_EXECUTION_REQUIRED"
    }
    adapter_keys = {
        key for key in deficits
        if classify(key)[0] == "MECHANICAL_ADAPTER_REQUIRED"
    }
    unresolved_keys = {
        key for key in deficits
        if classify(key)[0] == "ORIGIN_OR_FORMAT_UNRESOLVED"
    }

    # Estimate from recorded parent usage; candidate prompt delta is a proxy.
    estimates = []
    input_rate = batch["input_toman_per_million"]
    output_rate = batch["output_toman_per_million"]
    for key in sorted(fresh_keys):
        candidate_ids = sorted(candidates_by_request[key])
        usage = [
            candidates[cid].get("usage_as_reported") or {}
            for cid in candidate_ids
        ]
        values = [
            u["prompt_tokens"] for u in usage
            if type(u.get("prompt_tokens")) is int
        ]
        parent_input = max(values) if values else None
        old_prompt = (
            comparison["arms"]["reference"]["system_prompt_sha256"]
        )
        del old_prompt  # Identity is preserved in C240; no tokenizer assumed.

        if parent_input is None:
            estimated_input = None
        elif key[0] == "reference":
            estimated_input = parent_input
        else:
            # Disclosed planning reserve, not a tokenizer count or hard bound.
            estimated_input = (
                parent_input + len(policy["system_prompt"].encode("utf-8"))
            )
        estimates.append({
            "arm": key[0],
            "logical_request_sha256": key[1],
            "estimated_prompt_tokens": estimated_input,
            "maximum_requested_output_tokens": 600,
            "historical_estimate_toman": (
                (estimated_input * input_rate + 600 * output_rate) / 1_000_000
                if estimated_input is not None else None
            ),
        })

    all_costs_known = all(
        row["historical_estimate_toman"] is not None for row in estimates
    )
    summary = {
        "cell": "C241",
        "status": "DEFICIT_AUDIT_AND_COMMON_RUBRIC_PROPOSAL_COMPLETE",
        "C240_receipt": package_pin,
        "audited_deficit_count": len(audit),
        "category_counts": dict(Counter(row["category"] for row in audit)),
        "category_meanings": {
            "REUSABLE": "قابل‌مصرف در سطح درخواست و پاسخ؛ پذیرش Runtime رسمی نیست",
            "MECHANICAL_ADAPTER_REQUIRED": "قالب مکانیکی با حفظ بایت پاسخ باز شده؛ Adapter نیازمند پذیرش است",
            "ORIGIN_OR_FORMAT_UNRESOLVED": "نباید فقدان پاسخ یا نیاز قطعی به ارسال تلقی شود",
            "FRESH_EXECUTION_REQUIRED": "در Producerهای صریح کنترل‌شده، پاسخ کامل منطبق احراز نشد",
        },
        "audit": audit,
        "rubric_receipt": rubric_pin,
        "rubric_status_counts": dict(Counter(
            row["rubric_status"] for row in rubric
        )),
        "fresh_requests_in_enumerated_evidence_scope": len(fresh_keys),
        "adapter_requests_not_counted_as_fresh": len(adapter_keys),
        "unresolved_requests_not_counted_as_fresh": len(unresolved_keys),
        "repository_wide_absence_claimed": False,
        "cost_estimates": estimates,
        "historical_total_estimate_toman": (
            sum(row["historical_estimate_toman"] for row in estimates)
            if all_costs_known else None
        ),
        "tariff_source": batch_pin,
        "input_toman_per_million": input_rate,
        "output_toman_per_million": output_rate,
        "cost_method": "PARENT_REPORTED_PROMPT_USAGE_PLUS_CANDIDATE_PROMPT_BYTE_RESERVE",
        "current_tariff_verified": False,
        "estimate_is_hard_cap": False,
        "approved_future_seed": 20261027,
        "pool50_grouping_transferred_to_35_scenarios": False,
        "remaining_effective_decisions": [
            "APPROVE_COMMON_RUBRIC_OR_RESOLVE_BLOCKED_EXPECTATIONS",
            "APPROVE_MECHANICAL_ADAPTER_AND_DECLARED_PROVENANCE_SCOPE",
            "RESOLVE_UNKNOWN_ENVELOPES_BEFORE_TREATING_THEM_AS_MISSING",
            "AUTHORIZE_ONLY_FINAL_REQUIRED_REQUESTS_AFTER_RUBRIC_APPROVAL",
        ],
        "input_receipts": list(verified.values()),
        "execution_authorized": False,
        "new_api_calls": 0,
        "new_model_runs": 0,
        "new_ranking_runs": 0,
        "bootstrap_runs": 0,
        "historical_results_overwritten": False,
        "selected_version_gate": "BLOCKED",
    }
    output_pin = save("C241.R11_deficit_audit_and_rubric_decisions", summary)

    effective_rows = [{
        "regression_spec_id": row["regression_spec_id"],
        "expected_action": row["expected_action"],
        "status": row["rubric_status"],
        "unresolved_issues": row["unresolved_issues"],
    } for row in rubric]

    print(json.dumps({
        key: value for key, value in summary.items()
        if key not in {"audit", "input_receipts", "cost_estimates"}
    } | {
        "deficit_decision_table": [{
            "arm": row["arm"],
            "request": row["logical_request_sha256"],
            "category": row["category"],
            "reason": row["reason"],
            "evidence_findings": [{
                "source": item["source"],
                "state": item["state"],
                "reason": item.get("reason"),
                "adapter_steps": item.get("adapter_steps", []),
            } for item in row["observations"]],
        } for row in audit],
        "rubric_decision_table": effective_rows,
        "output": output_pin,
    }, ensure_ascii=False, indent=2))
    return summary


V3_C241 = run_v3_c241()
````

````python
# NB22E-C242 — Reuse existing witnesses and compute effective deficits.
# No network/generation/ranking/Bootstrap.

def run_v3_c242():
    root = Path(
        "/Users/macbookpro/AMIR_DATA/00_WORKS/01_PROJECTS/"
        "RAG-projects/RAG_finance"
    ).resolve()
    folder = root / "reports/phase2/quality_improvement/NB22E.lab.v3"

    def require(ok, message):
        if not ok:
            raise ValueError(message)

    def sha(raw):
        return hashlib.sha256(raw).hexdigest()

    def enc(value):
        return (
            json.dumps(value, ensure_ascii=False, sort_keys=True, indent=2)
            + "\n"
        ).encode("utf-8")

    def read_pin(pin):
        path = (root / pin["path"]).resolve()
        require(path.is_relative_to(root), "C242_PATH_OUTSIDE_PROJECT")
        require(path.is_file(), "C242_MISSING_FILE:" + str(path))
        raw = path.read_bytes()
        require(sha(raw) == pin["file_bytes_sha256"],
                "C242_HASH_MISMATCH:" + pin["path"])
        return json.loads(raw.decode("utf-8"))

    def load(stem, digest):
        pin = {
            "path": str((folder / f"{stem}.{digest}.json").relative_to(root)),
            "file_bytes_sha256": digest,
        }
        return read_pin(pin), pin

    def save(stem, value):
        raw = enc(value)
        digest = sha(raw)
        path = folder / f"{stem}.{digest}.json"
        if path.exists():
            require(path.read_bytes() == raw, "C242_OUTPUT_COLLISION")
        else:
            with path.open("xb") as handle:
                handle.write(raw)
                handle.flush()
                os.fsync(handle.fileno())
        require(sha(path.read_bytes()) == digest, "C242_RELOAD_FAILED")
        return {
            "path": str(path.relative_to(root)),
            "file_bytes_sha256": digest,
            "reload_validation": "PASS",
        }

    audit, audit_pin = load(
        "C241.R11_deficit_audit_and_rubric_decisions",
        "bae1ef5b5061c2d142c434d090795fd43b758d0f440e468ba75fd05468389672",
    )
    rubric = read_pin(audit["rubric_receipt"])
    plan = read_pin(audit["C240_receipt"])
    probe, probe_pin = load(
        "C200.R11_output_cap_probe_review",
        "1410eed10303881fd7fabec24575c582afff51a6c4adf3e15c1ff127af14f037",
    )
    packet, packet_pin = load(
        "C193.R11_batch_semantic_review_packet",
        "0b20b53e6812c69c47809a027f2c33c59e285bc6baf509db91d4e2707f460a4b",
    )
    candidates = {r["candidate_id"]: r for r in packet["candidate_reviews"]}
    probe_by_spec = {
        spec_id: row
        for row in probe["reviews"]
        for spec_id in row["scenario_ids"]
    }

    repaired = []
    revised_rows = []
    for original in rubric["rows"]:
        row = json.loads(json.dumps(original))
        spec_id = row["regression_spec_id"]

        if (
            spec_id in probe_by_spec
            and "CASE_SPECIFIC_CONTENT_WITNESS_NOT_BOUND"
            in row["unresolved_issues"]
        ):
            context = candidates[row["candidate_id"]]["exact_dispatched_context"]
            witnesses = probe_by_spec[spec_id]["witnesses"]
            require(bool(witnesses), "C242_PROBE_WITNESSES_MISSING")

            for witness in witnesses:
                quote = witness["exact_quote"]
                require(
                    context[witness["start"]:witness["end"]] == quote,
                    "C242_PROBE_CONTEXT_SPAN_MISMATCH:" + spec_id,
                )
                matches = list(re.finditer(
                    r"\[منبع " + str(witness["source_number"])
                    + r"\]\n(.*?)(?=\n\[منبع \d+\]\n|\Z)",
                    context, re.S,
                ))
                require(
                    len(matches) == 1
                    and quote in matches[0].group(1)
                    and ("شناسه چانک: " + witness["chunk_id"])
                    in matches[0].group(1),
                    "C242_PROBE_SOURCE_BINDING_MISMATCH:" + spec_id,
                )

            require(
                isinstance(row["required_evidence"], dict),
                "C242_EXPECTED_CONTENT_EVIDENCE_OBJECT",
            )
            row["required_evidence"]["additional_existing_witnesses"] = witnesses
            row["required_evidence"]["additional_witness_receipt"] = probe_pin
            row["unresolved_issues"].remove(
                "CASE_SPECIFIC_CONTENT_WITNESS_NOT_BOUND"
            )
            row["rubric_status"] = (
                "BLOCKED_EXPECTATION_OR_EVIDENCE"
                if row["unresolved_issues"]
                else "PROPOSED_FOR_REFERENCE_APPROVAL"
            )
            row["probe_behavior_verdict_transferred"] = False
            row["new_reference_approval_granted"] = False
            repaired.append(spec_id)

        revised_rows.append(row)

    require(len(repaired) == 2, "C242_EXPECTED_TWO_WITNESS_CONNECTIONS")
    rubric_by_spec = {r["regression_spec_id"]: r for r in revised_rows}
    comparison = plan["prospective_response_comparison"]

    scenario_ids_by_request = {}
    for row in comparison["population"]:
        for arm in ("reference", "candidate"):
            key = (arm, row[arm + "_logical_request_sha256"])
            scenario_ids_by_request.setdefault(key, set()).update(
                row["scenario_ids"]
            )

    # Read the precise unresolved raw fields without accepting new model aliases.
    identity_diagnostics = []
    for row in audit["audit"]:
        if row["category"] != "ORIGIN_OR_FORMAT_UNRESOLVED":
            continue
        for observation in row["observations"]:
            value = read_pin(observation["receipt"])
            if "choices" not in value:
                body = value.get("response_body")
                if isinstance(body, str):
                    value = json.loads(body)
                elif isinstance(body, dict):
                    value = body

            choices = value.get("choices", []) if isinstance(value, dict) else []
            choice = choices[0] if len(choices) == 1 else {}
            content = choice.get("message", {}).get("content")
            identity_diagnostics.append({
                "arm": row["arm"],
                "logical_request_sha256": row["logical_request_sha256"],
                "receipt": observation["receipt"],
                "recorded_response_model": value.get("model"),
                "finish_reason": choice.get("finish_reason"),
                "choice_count": len(choices),
                "content_type": type(content).__name__,
                "content_characters": len(content) if isinstance(content, str) else None,
                "text_hash": (
                    sha(content.encode("utf-8"))
                    if isinstance(content, str) else None
                ),
                "model_alias_accepted_by_this_cell": False,
                "decision": "HOLD_IDENTITY_QUALIFICATION_NO_RESEND",
            })

    decision_table = []
    effective_fresh = set()
    all_fresh = set()
    for row in audit["audit"]:
        key = (row["arm"], row["logical_request_sha256"])
        spec_ids = scenario_ids_by_request[key]
        proposed = sorted(
            spec_id for spec_id in spec_ids
            if rubric_by_spec[spec_id]["rubric_status"]
            == "PROPOSED_FOR_REFERENCE_APPROVAL"
        )
        blocked = sorted(set(spec_ids) - set(proposed))

        if row["category"] == "FRESH_EXECUTION_REQUIRED":
            all_fresh.add(key)
            if proposed:
                effective_fresh.add(key)

        decision_table.append({
            "arm": key[0],
            "logical_request_sha256": key[1],
            "mechanical_category": row["category"],
            "reason": row["reason"],
            "proposed_scenario_ids": proposed,
            "blocked_scenario_ids": blocked,
            "needed_if_proposed_rubric_is_approved": key in effective_fresh,
            "dispatch_authorized": False,
        })

    costs = {
        (row["arm"], row["logical_request_sha256"]): row
        for row in audit["cost_estimates"]
    }
    selected_estimates = [costs[key] for key in sorted(effective_fresh)]
    costs_known = all(
        row["historical_estimate_toman"] is not None
        for row in selected_estimates
    )

    revised_rubric = {
        **rubric,
        "schema_version": "NB22E_COMMON_POLICY_RUBRIC_PROPOSAL_V1_1",
        "parent_rubric_receipt": audit["rubric_receipt"],
        "change": "CONNECT_TWO_EXISTING_C200_WITNESSES_ONLY",
        "new_auxiliary_evidence": probe_pin,
        "rows": revised_rows,
        "acceptance_criteria_changed": False,
        "behavior_results_transferred_between_profiles": False,
        "approval_status": "PROPOSAL_NOT_OFFICIAL_REFERENCE",
    }
    rubric_pin = save(
        "C242.R11_common_policy_rubric_proposal_v1_1", revised_rubric
    )

    result = {
        "cell": "C242",
        "status": "EFFECTIVE_DEFICITS_AND_REFERENCE_DECISIONS_READY",
        "input_receipts": {
            "C241": audit_pin,
            "C200_existing_witnesses": probe_pin,
            "C193_fixed_context": packet_pin,
        },
        "revised_rubric_receipt": rubric_pin,
        "existing_witness_connections_added": repaired,
        "rubric_status_counts": dict(Counter(
            row["rubric_status"] for row in revised_rows
        )),
        "all_fresh_deficits_in_enumerated_records": len(all_fresh),
        "fresh_requests_covering_proposed_scenarios": len(effective_fresh),
        "fresh_requests_covering_only_blocked_scenarios":
            len(all_fresh - effective_fresh),
        "fresh_by_arm_for_proposed_scenarios": dict(Counter(
            key[0] for key in effective_fresh
        )),
        "identity_unresolved_requests": len({
            row["logical_request_sha256"] for row in identity_diagnostics
        }),
        "identity_diagnostics": identity_diagnostics,
        "request_decision_table": decision_table,
        "effective_request_cost_estimates": selected_estimates,
        "historical_effective_estimate_toman": (
            sum(row["historical_estimate_toman"] for row in selected_estimates)
            if costs_known else None
        ),
        "tariff_source": audit["tariff_source"],
        "cost_method": audit["cost_method"],
        "current_tariff_verified": False,
        "estimate_is_hard_cap": False,
        "effective_reference_decisions": [
            {
                "id": row["regression_spec_id"],
                "expected_action": row["expected_action"],
                "status": row["rubric_status"],
                "unresolved_issues": row["unresolved_issues"],
            } for row in revised_rows
            if row["rubric_status"] == "BLOCKED_EXPECTATION_OR_EVIDENCE"
        ],
        "approval_needed": [
            "APPROVE_COMMON_RUBRIC_FOR_ITS_EXPLICIT_SCENARIO_SCOPE",
            "RESOLVE_TWO_RAW_MODEL_IDENTITIES_WITHOUT_GUESSING_ALIASES",
            "AUTHORIZE_ONLY_EFFECTIVE_UNIQUE_REQUESTS_AFTER_APPROVAL",
        ],
        "new_api_calls": 0,
        "new_model_runs": 0,
        "new_ranking_runs": 0,
        "bootstrap_runs": 0,
        "gold_changes": 0,
        "execution_authorized": False,
        "historical_results_overwritten": False,
        "selected_version_gate": "BLOCKED",
    }
    output_pin = save("C242.R11_effective_execution_decisions", result)
    print(json.dumps({
        key: value for key, value in result.items()
        if key not in {"effective_request_cost_estimates"}
    } | {"output": output_pin}, ensure_ascii=False, indent=2))
    return result


V3_C242 = run_v3_c242()
````

````python
# NB22E-C245 — Offline response binding and review packet.
# No network, generation, retry, model-alias acceptance, or Bootstrap.

def run_v3_c245():
    root = Path(
        "/Users/macbookpro/AMIR_DATA/00_WORKS/01_PROJECTS/"
        "RAG-projects/RAG_finance"
    ).resolve()
    folder = root / "reports/phase2/quality_improvement/NB22E.lab.v3"

    def require(ok, message):
        if not ok:
            raise ValueError(message)

    def sha(raw):
        return hashlib.sha256(raw).hexdigest()

    def enc(value):
        return (
            json.dumps(value, ensure_ascii=False, sort_keys=True, indent=2)
            + "\n"
        ).encode("utf-8")

    def read_bytes(pin):
        path = (root / pin["path"]).resolve()
        require(path.is_relative_to(root), "C245_PATH_OUTSIDE_PROJECT")
        require(path.is_file(), "C245_MISSING_FILE:" + str(path))
        raw = path.read_bytes()
        require(
            sha(raw) == pin["file_bytes_sha256"],
            "C245_HASH_MISMATCH:" + pin["path"],
        )
        return raw

    def read_pin(pin):
        return json.loads(read_bytes(pin).decode("utf-8"))

    def load(stem, digest):
        pin = {
            "path": str((folder / f"{stem}.{digest}.json").relative_to(root)),
            "file_bytes_sha256": digest,
        }
        return read_pin(pin), pin

    def save(stem, value):
        raw = enc(value)
        digest = sha(raw)
        path = folder / f"{stem}.{digest}.json"
        if path.exists():
            require(path.read_bytes() == raw, "C245_OUTPUT_COLLISION")
        else:
            with path.open("xb") as handle:
                handle.write(raw)
                handle.flush()
                os.fsync(handle.fileno())
        require(sha(path.read_bytes()) == digest, "C245_RELOAD_FAILED")
        return {
            "path": str(path.relative_to(root)),
            "file_bytes_sha256": digest,
            "reload_validation": "PASS",
        }

    execution, execution_pin = load(
        "C244.R11_approved_policy_batch_execution",
        "54a105cfec43b5b7023f548248300ed4939829cde8a993d85db8b9694f0fccee",
    )
    plan = read_pin(execution["plan_receipt"])
    rubric = read_pin(execution["frozen_rubric_receipt"])
    packet, packet_pin = load(
        "C193.R11_batch_semantic_review_packet",
        "0b20b53e6812c69c47809a027f2c33c59e285bc6baf509db91d4e2707f460a4b",
    )

    rubric_by_id = {
        row["regression_spec_id"]: row for row in rubric["rows"]
    }
    candidates = {
        row["candidate_id"]: row for row in packet["candidate_reviews"]
    }
    planned = {
        (job["arm"], job["logical_request_sha256"]): job
        for job in plan["jobs"]
    }
    require(len(planned) == 11, "C245_EXPECTED_ELEVEN_JOBS")

    seen = set()
    review_rows = []
    models = Counter()

    for summary in execution["records"]:
        require(
            "record_receipt" in summary,
            "C245_RESULT_RECORD_MISSING_NO_RESEND",
        )
        record = read_pin(summary["record_receipt"])
        key = (record["arm"], record["logical_request_sha256"])
        require(
            key in planned and key not in seen,
            "C245_UNEXPECTED_OR_DUPLICATE_RESULT",
        )
        seen.add(key)
        job = planned[key]

        require(
            sha(enc(job["request"])) == key[1],
            "C245_LOGICAL_REQUEST_HASH_MISMATCH",
        )
        require(
            record["plan_receipt"] == execution["plan_receipt"]
            and record["frozen_rubric_receipt"]
            == execution["frozen_rubric_receipt"]
            and record["scenario_ids"] == job["scenario_ids"],
            "C245_PLAN_RESULT_BINDING_MISMATCH",
        )

        started = read_pin(record["started_receipt"])
        require(
            started["logical_request_sha256"] == key[1]
            and started["request"] == job["request"]
            and started["plan_receipt"] == execution["plan_receipt"]
            and started["frozen_rubric_receipt"]
            == execution["frozen_rubric_receipt"]
            and started["attempt_number"] == 1
            and started["scenario_ids"] == job["scenario_ids"]
            and started["prior_incomplete_response_receipts"]
            == job["prior_incomplete_response_receipts"],
            "C245_STARTED_RECEIPT_BINDING_MISMATCH",
        )

        response = read_pin(record["raw_response_receipt"])
        choices = response.get("choices", [])
        require(len(choices) == 1, "C245_RESPONSE_CHOICE_COUNT")
        answer = choices[0].get("message", {}).get("content")
        require(
            record["http_status"] == 200
            and choices[0].get("finish_reason") == "stop"
            and isinstance(answer, str) and bool(answer.strip())
            and answer == record["answer"]
            and sha(answer.encode("utf-8")) == record["answer_text_sha256"],
            "C245_COMPLETE_RESPONSE_BINDING_MISMATCH",
        )
        raw_model = response.get("model")
        require(
            isinstance(raw_model, str)
            and raw_model == record["response_model"],
            "C245_RAW_MODEL_FIELD_MISMATCH",
        )
        models[raw_model] += 1

        scenarios = []
        for spec_id in job["scenario_ids"]:
            row = rubric_by_id[spec_id]
            require(
                row["rubric_status"] == "APPROVED_FOR_LIMITED_POLICY_COMPARISON",
                "C245_BLOCKED_SCENARIO_IN_EXECUTED_QUEUE",
            )
            candidate = candidates[row["candidate_id"]]
            context = candidate["exact_dispatched_context"]
            user_messages = [
                message["content"] for message in job["request"]["messages"]
                if message["role"] == "user"
            ]
            require(
                any(context in text and row["question"] in text
                    for text in user_messages),
                "C245_QUESTION_CONTEXT_REQUEST_MISMATCH",
            )
            scenarios.append({
                "regression_spec_id": spec_id,
                "candidate_id": row["candidate_id"],
                "question": row["question"],
                "context_sha256": sha(context.encode("utf-8")),
                "exact_context": context,
                "expected_action": row["expected_action"],
                "common_acceptance_criteria":
                    row["common_acceptance_criteria"],
                "required_evidence": row["required_evidence"],
                "original_specification": row["original_specification"],
                "behavior_verdict": None,
            })

        review_rows.append({
            "arm": key[0],
            "logical_request_sha256": key[1],
            "requested_model": job["request"]["model"],
            "raw_reported_model": raw_model,
            "model_identity_status": "HOLD_PROVIDER_MAPPING_NOT_ESTABLISHED",
            "answer": answer,
            "answer_text_sha256": record["answer_text_sha256"],
            "usage_as_reported": response.get("usage"),
            "scenarios": scenarios,
            "record_receipt": summary["record_receipt"],
            "started_receipt": record["started_receipt"],
            "raw_response_receipt": record["raw_response_receipt"],
            "complete_response_binding_verified": True,
        })

    require(seen == set(planned), "C245_RESULT_COVERAGE_MISMATCH")

    review_pin = save("C245.R11_identity_held_response_review_packet", {
        "cell": "C245",
        "execution_receipt": execution_pin,
        "plan_receipt": execution["plan_receipt"],
        "frozen_rubric_receipt": execution["frozen_rubric_receipt"],
        "fixed_context_receipt": packet_pin,
        "common_rules": rubric["rules"],
        "scope": "CONTENT_REVIEW_POSSIBLE_WITH_MODEL_IDENTITY_HOLD_PRESERVED",
        "review_rows": review_rows,
        "model_equivalence_granted": False,
        "comparison_acceptance_granted": False,
    })

    result = {
        "cell": "C245",
        "status": "ELEVEN_COMPLETE_RESPONSES_BOUND_MODEL_IDENTITY_HOLD_PRESERVED",
        "review_packet": review_pin,
        "verified_complete_request_count": len(review_rows),
        "covered_scenario_count": len({
            row["regression_spec_id"]
            for item in review_rows for row in item["scenarios"]
        }),
        "requested_model": "DeepSeek-V4-Flash",
        "raw_reported_model_counts": dict(models),
        "model_equivalence_granted": False,
        "provider_mapping_required": (
            "DOCUMENTED_MAPPING_OF_THE_EXACT_REPORTED_MODEL_IDENTIFIER_"
            "TO_THE_REQUESTED_MODEL_FOR_THE_USED_GATEWAY"
        ),
        "new_api_calls": 0,
        "automatic_retries": 0,
        "new_behavior_passes": 0,
        "comparison_acceptance_granted": False,
        "selected_version_gate": "BLOCKED",
        "answers_for_review": [{
            "arm": item["arm"],
            "raw_reported_model": item["raw_reported_model"],
            "answer": item["answer"],
            "answer_text_sha256": item["answer_text_sha256"],
            "scenarios": [{
                key: scenario[key]
                for key in (
                    "regression_spec_id", "question", "expected_action",
                    "common_acceptance_criteria",
                )
            } for scenario in item["scenarios"]],
        } for item in review_rows],
    }
    output_pin = save("C245.R11_response_binding_and_model_identity", result)
    print(json.dumps(
        result | {"output": output_pin}, ensure_ascii=False, indent=2
    ))
    return result


V3_C245 = run_v3_c245()
````

````python
# NB22E-C246 — Corrected standalone scoped review.
# No API, retry, generation, ranking, or Bootstrap.

def run_v3_c246():
    root = Path(
        "/Users/macbookpro/AMIR_DATA/00_WORKS/01_PROJECTS/"
        "RAG-projects/RAG_finance"
    ).resolve()
    folder = root / "reports/phase2/quality_improvement/NB22E.lab.v3"

    def require(ok, message):
        if not ok:
            raise ValueError(message)

    def sha(raw):
        return hashlib.sha256(raw).hexdigest()

    def enc(value):
        return (
            json.dumps(value, ensure_ascii=False, sort_keys=True, indent=2)
            + "\n"
        ).encode("utf-8")

    def read_pin(pin):
        path = (root / pin["path"]).resolve()
        require(path.is_relative_to(root), "C246_PATH_OUTSIDE_PROJECT")
        require(path.is_file(), "C246_MISSING_FILE:" + str(path))
        raw = path.read_bytes()
        require(
            sha(raw) == pin["file_bytes_sha256"],
            "C246_HASH_MISMATCH:" + pin["path"],
        )
        return json.loads(raw.decode("utf-8"))

    packet_pin = {
        "path": str((folder / (
            "C245.R11_identity_held_response_review_packet."
            "54a825f4ddf442c67f714d47435dd0003dae16b526740a9fd6af795de059c6e5.json"
        )).relative_to(root)),
        "file_bytes_sha256":
            "54a825f4ddf442c67f714d47435dd0003dae16b526740a9fd6af795de059c6e5",
    }
    packet = read_pin(packet_pin)
    rubric = read_pin(packet["frozen_rubric_receipt"])
    rubric_by_id = {row["regression_spec_id"]: row for row in rubric["rows"]}

    def witness(context, number, quote):
        matches = list(re.finditer(
            r"\[منبع " + str(number)
            + r"\]\n(.*?)(?=\n\[منبع \d+\]\n|\Z)",
            context, re.S,
        ))
        require(len(matches) == 1, "C246_SOURCE_NOT_UNIQUE")
        match = matches[0]
        block = match.group(1)
        require(
            block.count(quote) == 1,
            f"C246_QUOTE_MISMATCH:{number}:{quote}",
        )
        chunk = re.search(r"شناسه چانک: ([^\n]+)", block)
        require(chunk is not None, "C246_CHUNK_ID_MISSING")
        start = match.start(1) + block.index(quote)
        return {
            "source_number": number,
            "chunk_id": chunk.group(1),
            "exact_quote": quote,
            "start": start,
            "end": start + len(quote),
            "span_unit": "UNICODE_CODE_POINT",
            "offset_reference": "EXACT_CONTEXT_IN_C245_REVIEW_PACKET",
        }

    decisions = {
        "REG_0281e0049a2f0f23b738ed3e": {
            "hashes": {
                "candidate": "815053f371f3f5514420fcf50ad7634583f7d80c2ee9d75882770b95b929502b",
                "reference": "9c5205cca6d5aa41d6edc1b020dc9292e02389fd6af634c14007a8290758f105",
            },
            "finding": "CORE_LEARNING_AND_REVERSIBLE_ERROR_LIMIT_SUPPORTED",
            "reason": (
                "هر دو پاسخ، یادگیری از اشتباه را به تجربهٔ امن و "
                "جبران‌پذیر مقید کرده‌اند؛ ادعاهای جزئی افزوده "
                "در این کنترل کامل داوری نمی‌شوند."
            ),
            "quotes": [
                (3, "اما آزمایش‌هایی که یادگیری داره، جبران‌پذیره،"),
            ],
            "failure": False,
        },
        "REG_ca2c12cf8a09e0dbcb158e24": {
            "hashes": {
                "candidate": "29b8b84fc4abd3a825ff06964f8a65e9ebe3ecb98f771db18dee930e709c34b5",
            },
            "finding": "CORE_EARNING_AS_FINANCIAL_LITERACY_DOMAIN_SUPPORTED",
            "reason": "معرفی کسب درآمد به‌عنوان بخشی از سواد مالی در شاهد صریح وجود دارد.",
            "quotes": [
                (1, "سواد مالی درباره بخشی از رفتارهای قابل مدیریت ماست: کسب درآمد، هزینه، پس‌انداز، بدهی، سرمایه‌گذاری و تصمیم‌گیری."),
            ],
            "failure": False,
        },
        "REG_b8f1cb382c66e6fc76989363": {
            "hashes": {
                "candidate": "05c20ac037ade0d358a30a5ccce6f019eea1ee8ccb639351386d0a7dcded0ef3",
            },
            "finding": "SUPPORTED_SELF_ASSESSMENT_INTERPRETATION_OMITTED",
            "reason": (
                "پاسخ سؤال را به تعریف عدد ریاضی محدود کرده؛ "
                "شاهد معنای عدد یک در خودارزیابی را بیان می‌کند. "
                "پاسخ مقید مصوب ارائه نشده است."
            ),
            "quotes": [
                (5, "اگر هر عبارت در مورد شما **به‌ندرت** صادقه، عدد **یک** رو در نظر بگیرین."),
            ],
            "failure": True,
        },
        "REG_41bf5ee8206b3fe9454ee6e3": {
            "hashes": {
                "candidate": "f87d5223866ab68454f0fde8d6546b3ac5eb005a9ac368d68989ea1db286d23e",
            },
            "finding": "CORE_TRANSACTION_RISK_SUPPORTED_ADDITIONAL_SCOPE_REVIEW_PENDING",
            "reason": (
                "بخش جسارت و اقدام جوانان در تراکنش‌ها پشتیبانی "
                "دارد؛ دامنهٔ آمارها و استدلال‌های اضافی هنوز "
                "کامل داوری نشده‌اند."
            ),
            "quotes": [
                (1, "جوان‌ها معمولاً جسورترن و زودتر اقدام می‌کنن، در نتیجه در معرضش قرار می‌گیرن."),
            ],
            "failure": False,
        },
        "REG_859ea88832af3714996a7dc4": {
            "hashes": {
                "candidate": "8d9d54e0df10ce30d3af9560dbcc285ddd9406a42dc01a3fa0542106f564a1fd",
            },
            "finding": "CORE_VOLUNTEERING_BENEFITS_SUPPORTED",
            "reason": (
                "بی‌درآمدبودن کار داوطلبانه و نقش آن در ورود به کار "
                "و ارتباطات شغلی در شاهد آمده؛ تمام موارد فهرست "
                "پاسخ اینجا داوری نشده‌اند."
            ),
            "quotes": [
                (6, "کاری که درآمد ندارد؛ اما مزایای فراوانی دارد:"),
                (6, "راه‌های ورود به کار باز شده و ارتباطات شغلی شکل می‌گیرد."),
            ],
            "failure": False,
        },
        "REG_0e09b08de27d2e4b4a5f6e3b": {
            "hashes": {
                "candidate": "a74c08b91c359c6953c61a4cdd1952e8326abc7703590deac9529a13fda2db7a",
            },
            "finding": "CORE_FINANCIAL_LITERACY_DISTINCTIONS_SUPPORTED",
            "reason": (
                "تمایز سواد مالی از کسب‌وکار، مقدار پول، رشتهٔ "
                "دانشگاهی و موفقیت در شواهد موجود است."
            ),
            "quotes": [
                (1, "پس سواد مالی کارآفرینی نیست و کسب‌وکار هم نیست."),
                (2, "سواد مالی درباره میزان پول نیست؛ درباره مدیریت پول است."),
                (6, "سواد مالی رشته دانشگاهی نیست."),
                (8, "سواد مالی «سواد موفقیت» هم نیست."),
            ],
            "failure": False,
        },
        "REG_990035306ebed83cb7c49982": {
            "hashes": {
                "candidate": "4f4dab1a4b42edfd9b2ae1c9fdde4f9259c54bdb517a811edb0e83db3acbf028",
            },
            "finding": "CORE_INCOME_ALLOCATION_RECOMMENDATION_SUPPORTED",
            "reason": (
                "توصیهٔ تقسیم درآمد در شاهد وجود دارد؛ دیگر "
                "توصیه‌ها و پیوندهای پاسخ در این کنترل هسته‌ای "
                "کامل داوری نشده‌اند."
            ),
            "quotes": [
                (4, "در توصیه‌ای دیگر درآمد را میان چند مصرف تقسیم می‌کند: هزینه‌های ضروری زندگی، ذخیره برای روزهای بیماری و ناتوانی و بخشی برای سرمایه‌گذاری و حفظ ارزش مال."),
            ],
            "failure": False,
        },
        "REG_4e532bd5d37ed5e8579647cb": {
            "hashes": {
                "candidate": "f1c3ef8b14a009da3609fbc40b8aff31061f87b188d8b8976ba4e7270aaf8668",
            },
            "finding": "SOURCE_POSSIBILITY_STRENGTHENED_TO_IDENTITY",
            "reason": (
                "پاسخ می‌گوید فاصلهٔ درآمد و هزینه ثروت خالص فرد "
                "است؛ شاهد می‌گوید «می‌تونه ثروت خالص شما باشه». "
                "قید احتمال حفظ نشده؛ داوری فقط دربارهٔ وفاداری "
                "به همین منبع است."
            ),
            "quotes": [
                (5, "این فاصله بین درآمد و هزینه می‌تونه ثروت خالص شما باشه"),
            ],
            "failure": True,
        },
    }

    absent_specs = {
        "REG_3d322038d49eef122d5c1f43",
        "REG_90711f8448cc834242f9b636",
    }
    personal_specs = {
        "REG_387903cf3570e3e378731d21",
        "REG_2b38170afb73381b47607627",
    }
    company_spec = "REG_d9ba26cb4167a891bf05d07b"
    reviews, seen = [], set()
    unique_links = set()

    for item in packet["review_rows"]:
        answer = item["answer"]
        require(
            sha(answer.encode("utf-8")) == item["answer_text_sha256"],
            "C246_ANSWER_HASH_MISMATCH",
        )

        for scenario in item["scenarios"]:
            spec_id = scenario["regression_spec_id"]
            key = (item["arm"], spec_id)
            require(key not in seen, "C246_DUPLICATE_ARM_SCENARIO")
            seen.add(key)
            context = scenario["exact_context"]

            require(
                sha(context.encode("utf-8")) == scenario["context_sha256"],
                "C246_CONTEXT_HASH_MISMATCH",
            )
            require(
                scenario["common_acceptance_criteria"]
                == rubric_by_id[spec_id]["common_acceptance_criteria"],
                "C246_FROZEN_CRITERIA_CHANGED",
            )

            row = {
                "arm": item["arm"],
                "regression_spec_id": spec_id,
                "answer_text_sha256": item["answer_text_sha256"],
                "context_sha256": scenario["context_sha256"],
                "frozen_criteria": scenario["common_acceptance_criteria"],
                "raw_reported_model": item["raw_reported_model"],
                "model_identity_status": item["model_identity_status"],
                "full_response_pass_granted": False,
                "comparison_eligible": False,
            }

            if spec_id in decisions:
                decision = decisions[spec_id]
                require(
                    item["answer_text_sha256"]
                    == decision["hashes"][item["arm"]],
                    "C246_REVIEWED_ANSWER_CHANGED",
                )
                row.update({
                    "finding": decision["finding"],
                    "reason_fa": decision["reason"],
                    "witnesses": [
                        witness(context, number, quote)
                        for number, quote in decision["quotes"]
                    ],
                    "content_disposition": (
                        "FAIL_CONTENT_CRITERION_ESTABLISHED"
                        if decision["failure"] else
                        "CORE_SUPPORTED_FULL_RESPONSE_REVIEW_PENDING"
                    ),
                    "additional_claim_review": "NOT_COMPLETE",
                })

            elif spec_id in absent_specs:
                evidence_id = scenario["original_specification"][
                    "historical_link_expectation"
                ]["evidence_id"]
                require(
                    ("شناسه چانک: " + evidence_id) not in context,
                    "C246_EXPECTED_HISTORICAL_SOURCE_ABSENCE_CHANGED",
                )
                row.update({
                    "finding": "HISTORICAL_SELECTED_SOURCE_ABSENT",
                    "content_disposition": "NOT_APPLICABLE_NOT_A_CITATION_PASS",
                    "historical_evidence_id": evidence_id,
                    "witnesses": [],
                    "reason_fa": (
                        "منبع منتخب تاریخی در زمینهٔ فعلی نیست؛ "
                        "این نتیجه قبولی استنادهای فعلی نیست."
                    ),
                })

            else:
                if spec_id in personal_specs:
                    require(
                        item["answer_text_sha256"]
                        == "e55cb4b5469c1c96f2d7520352dc49418ce4ca8a2b3589e986c4dcaf20d1533b",
                        "C246_PERSONAL_FINANCE_ANSWER_CHANGED",
                    )
                    claim = "ترجمه آن در زندگی هر فرد میتواند متفاوت باشد"
                    marker, number = "منبع 2 و 9", 9
                    quote = (
                        "آموزه‌هایی که اینجا می‌گیم نسخه کلیه؛ "
                        "ترجمه اون نسخه در زندگی شما می‌تونه متفاوت باشه."
                    )
                else:
                    require(
                        spec_id == company_spec
                        and item["answer_text_sha256"]
                        == "fa5af1145ad9be7a2476fe03725978ee245b460b539701c9923a9fd5ecb4500e",
                        "C246_UNEXPECTED_CITATION_CASE",
                    )
                    claim = (
                        "هیئت مدیره و مدیرعامل با اختیارات معین، "
                        "شرکت را مدیریت کرده و در چارچوب شرح وظایف "
                        "خود تصمیم‌گیری می‌کنند"
                    )
                    marker, number = "منبع 1 و 2", 1
                    quote = (
                        "با اختیارات معین، شرکت را مدیریت کرده و "
                        "در چارچوب شرح وظایف خود تصمیم‌گیری می‌کنند."
                    )

                require(
                    answer.count(claim) == 1 and marker in answer,
                    "C246_CURRENT_CLAIM_OR_CITATION_NOT_FOUND",
                )
                source_witness = witness(context, number, quote)
                historical_id = scenario["original_specification"][
                    "historical_link_expectation"
                ]["evidence_id"]
                require(
                    source_witness["chunk_id"] == historical_id,
                    "C246_SELECTED_SOURCE_CHANGED",
                )

                start = answer.index(claim)
                link_id = sha(enc({
                    "answer_sha": item["answer_text_sha256"],
                    "claim_start": start,
                    "claim_end": start + len(claim),
                    "chunk_id": source_witness["chunk_id"],
                }))
                unique_links.add(link_id)
                row.update({
                    "finding": "ONE_CURRENT_SUBCLAIM_LINK_TEXTUALLY_SUPPORTED",
                    "content_disposition": "SCOPED_LINK_SUPPORT_ONLY",
                    "current_claim": claim,
                    "claim_start": start,
                    "claim_end": start + len(claim),
                    "citation_marker": marker,
                    "current_link_identity": link_id,
                    "witnesses": [source_witness],
                    "historical_compound_claim_reused_without_reassessment": False,
                    "reason_fa": (
                        "پیوند همین زیرادعای فعلی به شاهد بررسی شد؛ "
                        "پشتیبانی به کل ادعای مرکب تاریخی یا سایر "
                        "استنادهای پاسخ تعمیم داده نمی‌شود."
                    ),
                    "other_claim_links_review": "NOT_COMPLETE",
                })

            reviews.append(row)

    require(len(reviews) == 14, "C246_EXPECTED_14_ARM_SCENARIO_ROWS")
    result = {
        "cell": "C246",
        "status": "SCOPED_CONTENT_AND_CURRENT_LINK_FINDINGS_RECORDED",
        "review_packet_receipt": packet_pin,
        "frozen_rubric_receipt": packet["frozen_rubric_receipt"],
        "review_method": "ASSISTANT_REVIEW_BOUND_TO_EXACT_TEXT_AND_SPANS",
        "independent_blind_review_claimed": False,
        "arm_scenario_row_count": len(reviews),
        "distinct_scenario_count": len({
            row["regression_spec_id"] for row in reviews
        }),
        "disposition_counts": dict(Counter(
            row["content_disposition"] for row in reviews
        )),
        "unique_current_subclaim_links_with_textual_support": len(unique_links),
        "duplicate_scenarios_counted_as_duplicate_links": False,
        "reviews": reviews,
        "full_response_passes_granted": 0,
        "all_additional_claims_adjudicated": False,
        "model_equivalence_granted": False,
        "model_identity_hold_preserved": True,
        "paired_comparison_acceptance_granted": False,
        "new_api_calls": 0,
        "new_ranking_runs": 0,
        "bootstrap_runs": 0,
        "prompt_changed": False,
        "historical_results_overwritten": False,
        "selected_version_gate": "BLOCKED",
    }

    raw = enc(result)
    digest = sha(raw)
    path = folder / f"C246.R11_scoped_content_and_link_review.{digest}.json"
    if path.exists():
        require(path.read_bytes() == raw, "C246_OUTPUT_COLLISION")
    else:
        with path.open("xb") as handle:
            handle.write(raw)
            handle.flush()
            os.fsync(handle.fileno())

    require(sha(path.read_bytes()) == digest, "C246_RELOAD_FAILED")
    output_pin = {
        "path": str(path.relative_to(root)),
        "file_bytes_sha256": digest,
        "reload_validation": "PASS",
    }
    print(json.dumps({
        key: value for key, value in result.items() if key != "reviews"
    } | {
        "decision_table": [{
            "arm": row["arm"],
            "regression_spec_id": row["regression_spec_id"],
            "disposition": row["content_disposition"],
            "reason_fa": row["reason_fa"],
        } for row in reviews],
        "output": output_pin,
    }, ensure_ascii=False, indent=2))
    return result


V3_C246 = run_v3_c246()
````

````python
# NB22E-C248 — Paired coverage under the frozen common rubric.
# No API, retry, generation, ranking, or Bootstrap.

def run_v3_c248():
    root = Path(
        "/Users/macbookpro/AMIR_DATA/00_WORKS/01_PROJECTS/"
        "RAG-projects/RAG_finance"
    ).resolve()
    folder = root / "reports/phase2/quality_improvement/NB22E.lab.v3"

    def require(ok, message):
        if not ok:
            raise ValueError(message)

    def sha(raw):
        return hashlib.sha256(raw).hexdigest()

    def enc(value):
        return (
            json.dumps(value, ensure_ascii=False, sort_keys=True, indent=2)
            + "\n"
        ).encode("utf-8")

    def read_raw(pin):
        path = (root / pin["path"]).resolve()
        require(path.is_relative_to(root), "C248_PATH_OUTSIDE_PROJECT")
        require(path.is_file(), "C248_MISSING_FILE:" + str(path))
        raw = path.read_bytes()
        require(sha(raw) == pin["file_bytes_sha256"],
                "C248_HASH_MISMATCH:" + pin["path"])
        return raw

    def read_pin(pin):
        return json.loads(read_raw(pin).decode("utf-8"))

    def load(stem, digest):
        pin = {
            "path": str((folder / f"{stem}.{digest}.json").relative_to(root)),
            "file_bytes_sha256": digest,
        }
        return read_pin(pin), pin

    plan, plan_pin = load(
        "C240.R11_prospective_comparison_and_test_groups",
        "3f68803a338bc43597de67732e6895f96267eebe7b4942f23885a4059022a108",
    )
    executed, executed_pin = load(
        "C244.R11_approved_policy_batch_execution",
        "54a105cfec43b5b7023f548248300ed4939829cde8a993d85db8b9694f0fccee",
    )
    reviewed, reviewed_pin = load(
        "C247.R11_completed_content_reviews",
        "605414bdb865113cf6c055d7eace3d2f769bbba0de1ea8b3bf1a8e7256d36b9a",
    )
    packet, packet_pin = load(
        "C193.R11_batch_semantic_review_packet",
        "0b20b53e6812c69c47809a027f2c33c59e285bc6baf509db91d4e2707f460a4b",
    )
    rubric = read_pin(executed["frozen_rubric_receipt"])
    approved = {
        row["regression_spec_id"]: row for row in rubric["rows"]
        if row["rubric_status"] == "APPROVED_FOR_LIMITED_POLICY_COMPARISON"
    }
    require(len(approved) == 28, "C248_EXPECTED_28_APPROVED_SCENARIOS")

    comparison = plan["prospective_response_comparison"]
    policy = read_pin(comparison["arms"]["candidate"]["policy_receipt"])
    candidates = {
        row["candidate_id"]: row for row in packet["candidate_reviews"]
    }

    requests, candidate_requests, scenario_requests = {}, {}, {}
    for row in comparison["population"]:
        parent = read_pin(row["parent_job_receipt"])
        for arm in ("reference", "candidate"):
            request = json.loads(json.dumps(parent["request"]))
            if arm == "candidate":
                systems = [
                    m for m in request["messages"] if m["role"] == "system"
                ]
                require(len(systems) == 1, "C248_SYSTEM_COUNT")
                systems[0]["content"] = policy["system_prompt"]
            identity = sha(enc(request))
            require(
                identity == row[arm + "_logical_request_sha256"],
                "C248_REQUEST_CHANGED",
            )
            key = (arm, identity)
            requests[key] = request
            candidate_requests[(arm, row["candidate_id"])] = key
            for spec_id in row["scenario_ids"]:
                if spec_id in approved:
                    scenario_requests[(arm, spec_id)] = key

    def unwrap(pin):
        value = read_pin(pin)
        for _ in range(5):
            if isinstance(value, dict) and "choices" in value:
                return value
            require(isinstance(value, dict), "C248_RESPONSE_ENVELOPE_NOT_OBJECT")
            if isinstance(value.get("raw_body_receipt"), dict):
                value = read_pin(value["raw_body_receipt"])
                continue
            possibilities = []
            for name in ("response_body", "body", "response", "raw_response"):
                item = value.get(name)
                if isinstance(item, str):
                    try:
                        item = json.loads(item)
                    except (ValueError, TypeError):
                        continue
                if isinstance(item, dict) and "choices" in item:
                    possibilities.append(item)
            require(len(possibilities) == 1,
                    "C248_RESPONSE_ENVELOPE_UNRESOLVED:" + pin["path"])
            value = possibilities[0]
        raise ValueError("C248_RESPONSE_ENVELOPE_DEPTH_LIMIT")

    observations = {key: [] for key in requests}

    def observe(key, pin, source):
        if key not in observations:
            return
        response = unwrap(pin)
        choices = response.get("choices", [])
        choice = choices[0] if len(choices) == 1 else {}
        answer = choice.get("message", {}).get("content")
        model = response.get("model")
        complete = (
            len(choices) == 1
            and choice.get("finish_reason") == "stop"
            and isinstance(answer, str) and bool(answer.strip())
        )
        observations[key].append({
            "source": source,
            "response_receipt": pin,
            "complete_text": complete,
            "finish_reason": choice.get("finish_reason"),
            "raw_reported_model": model,
            "model_name_consistency": (
                "REQUESTED_NAME_MATCH_CASE_INSENSITIVE"
                if isinstance(model, str)
                and model.lower() == requests[key]["model"].lower()
                else "HOLD_NO_PROVIDER_MAPPING"
            ),
            "answer": answer if isinstance(answer, str) else None,
            "answer_text_sha256": (
                sha(answer.encode("utf-8")) if isinstance(answer, str) else None
            ),
        })

    for candidate_id, row in candidates.items():
        observe(
            candidate_requests[("reference", candidate_id)],
            row["response_receipt"], "C192/C193",
        )

    for stem, digest in [
        (
            "C224.R11_seen_clarification_execution",
            "fd254f8e4822bd9541fdc4c26009a965d2b4f164d6d963528908feb6a4ce9220",
        ),
        (
            "C226.R11_seen_fidelity_execution",
            "2eabe09e4ef0abf7a918434f4d0c5782c3f362b8ca9f4880c72d4dc498718207",
        ),
    ]:
        document, _ = load(stem, digest)
        for record in document["candidate_results"]:
            job = read_pin(record["job_receipt"])
            observe(
                ("candidate", sha(enc(job["request"]))),
                record["response_receipt"], stem,
            )

    c233, _ = load(
        "C233.R11_six_case_policy_execution",
        "ff9988dcf4d3a002dc003d081491bf6bd395f998e47e431fdf43a80c3db54750",
    )
    p233 = read_pin(c233["plan_receipt"])
    jobs233 = {j["request_payload_sha256"]: j for j in p233["jobs"]}
    for record in c233["results"]:
        request = jobs233[record["request_payload_sha256"]]["request"]
        observe(
            ("candidate", sha(enc(request))),
            record["raw_response_receipt"], "C233",
        )

    for record in executed["records"]:
        observe(
            (record["arm"], record["logical_request_sha256"]),
            record["raw_response_receipt"], "C244",
        )

    known_reviews = {
        (row["arm"], row["regression_spec_id"], row["answer_text_sha256"]): row
        for row in reviewed["reviews"]
    }

    table = []
    for spec_id, criterion in sorted(approved.items()):
        for arm in ("reference", "candidate"):
            key = scenario_requests[(arm, spec_id)]
            complete = [
                item for item in observations[key] if item["complete_text"]
            ]
            hashes = {item["answer_text_sha256"] for item in complete}
            row = {
                "regression_spec_id": spec_id,
                "candidate_id": criterion["candidate_id"],
                "arm": arm,
                "logical_request_sha256": key[1],
                "expected_action": criterion["expected_action"],
                "frozen_criteria": criterion["common_acceptance_criteria"],
                "rubric_receipt": executed["frozen_rubric_receipt"],
                "observation_count": len(observations[key]),
                "comparison_acceptance_granted": False,
            }

            if not complete:
                row.update({
                    "response_status": "NO_COMPLETE_RESPONSE_IN_BOUND_INPUTS",
                    "content_status": "NOT_ASSESSABLE_NO_COMPLETE_RESPONSE",
                    "model_status": "NOT_ESTABLISHED",
                })
            elif len(hashes) != 1:
                row.update({
                    "response_status": "MULTIPLE_DISTINCT_COMPLETE_RESPONSES",
                    "content_status": "HOLD_PREDECLARE_RESPONSE_SELECTION",
                    "model_status": "NOT_COLLAPSED",
                    "distinct_answer_hashes": sorted(hashes),
                })
            else:
                # Identical text is retained with all receipts, not cherry-picked.
                answer_hash = next(iter(hashes))
                answer = complete[0]["answer"]
                known = known_reviews.get((arm, spec_id, answer_hash))
                has_identity_hold = any(
                    item["model_name_consistency"] == "HOLD_NO_PROVIDER_MAPPING"
                    for item in complete
                )
                row.update({
                    "response_status": "COMPLETE_TEXT_BOUND",
                    "answer_text_sha256": answer_hash,
                    "response_receipts": [
                        item["response_receipt"] for item in complete
                    ],
                    "raw_reported_models": sorted({
                        str(item["raw_reported_model"]) for item in complete
                    }),
                    "model_status": (
                        "HOLD_NO_PROVIDER_MAPPING" if has_identity_hold
                        else "REQUEST_RESPONSE_NAME_CONSISTENCY_ONLY"
                    ),
                })

                if known is not None:
                    row["content_status"] = known["content_disposition"]
                    row["content_review_receipt"] = reviewed_pin
                    row["reason_fa"] = known["reason_fa"]
                elif (
                    answer.strip() == "در منابع موجود پاسخ کافی پیدا نشد."
                    and criterion["expected_action"] in {
                        "ASK_RELEVANT_CLARIFICATION",
                        "ANSWER_SUPPORTED",
                        "BOUNDED_ANSWER_SUPPORTED",
                    }
                ):
                    row.update({
                        "content_status":
                            "FAIL_FROZEN_ACTION_CRITERION_BARE_ABSTENTION",
                        "reason_fa": (
                            "پاسخ فقط امتناع ثابت است؛ اقدام مصوبِ "
                            "پاسخ‌دادن یا توضیح‌خواهی را انجام نمی‌دهد."
                        ),
                        "review_scope": "EXACT_TEXT_ACTION_CHECK_ONLY",
                    })
                elif (
                    criterion["expected_action"] == "ASK_RELEVANT_CLARIFICATION"
                    and "در منابع موجود پاسخ کافی پیدا نشد." in answer
                ):
                    row.update({
                        "content_status":
                            "FAIL_FROZEN_AMBIGUITY_NO_ABSTENTION_CRITERION",
                        "reason_fa": (
                            "در حالت ابهام مؤثر، عبارت امتناع ثابت "
                            "برخلاف قاعدهٔ مصوب در پاسخ وجود دارد."
                        ),
                        "review_scope": "EXACT_FORBIDDEN_PHRASE_CHECK_ONLY",
                    })
                else:
                    row.update({
                        "content_status": "PENDING_COMMON_RUBRIC_CONTENT_REVIEW",
                        "answer_for_review": answer,
                        "reason_fa": (
                            "وجود پاسخ کامل خودکار قبولی نیست؛ "
                            "داوری مشترکِ متصل به این پاسخ هنوز "
                            "از رسیدهای مصرف‌شده احراز نشده است."
                        ),
                    })
            table.append(row)

    require(len(table) == 56, "C248_EXPECTED_56_ARM_SCENARIO_ROWS")
    pending = [
        row for row in table
        if row["content_status"] == "PENDING_COMMON_RUBRIC_CONTENT_REVIEW"
    ]

    result = {
        "cell": "C248",
        "status": "APPROVED_POPULATION_PAIRED_COVERAGE_AND_REVIEW_GAPS_BOUND",
        "frozen_rubric_receipt": executed["frozen_rubric_receipt"],
        "comparison_plan_receipt": plan_pin,
        "C244_execution_receipt": executed_pin,
        "C247_review_receipt": reviewed_pin,
        "historical_packet_receipt": packet_pin,
        "approved_scenario_count": 28,
        "arm_scenario_row_count": len(table),
        "unique_arm_request_count": len({
            (row["arm"], row["logical_request_sha256"]) for row in table
        }),
        "response_status_counts": dict(Counter(
            row["response_status"] for row in table
        )),
        "content_status_counts": dict(Counter(
            row["content_status"] for row in table
        )),
        "model_status_counts": dict(Counter(
            row["model_status"] for row in table
        )),
        "paired_rows": table,
        "remaining_common_rubric_review_count": len(pending),
        "model_alias_equivalence_granted": False,
        "other_scenarios_blocked_by_unrelated_model_cases": False,
        "population_pass_rate_computed": False,
        "formal_nonregression_granted": False,
        "new_api_calls": 0,
        "new_ranking_runs": 0,
        "bootstrap_runs": 0,
        "prompt_changed": False,
        "historical_results_overwritten": False,
        "selected_version_gate": "BLOCKED",
    }

    raw = enc(result)
    digest = sha(raw)
    path = folder / f"C248.R11_paired_population_review_matrix.{digest}.json"
    if path.exists():
        require(path.read_bytes() == raw, "C248_OUTPUT_COLLISION")
    else:
        with path.open("xb") as handle:
            handle.write(raw)
            handle.flush()
            os.fsync(handle.fileno())
    require(sha(path.read_bytes()) == digest, "C248_RELOAD_FAILED")

    print(json.dumps({
        key: value for key, value in result.items() if key != "paired_rows"
    } | {
        "pending_content_reviews": [{
            "arm": row["arm"],
            "regression_spec_id": row["regression_spec_id"],
            "expected_action": row["expected_action"],
            "answer_text_sha256": row["answer_text_sha256"],
            "answer": row["answer_for_review"],
            "model_status": row["model_status"],
        } for row in pending],
        "output": {
            "path": str(path.relative_to(root)),
            "file_bytes_sha256": digest,
            "reload_validation": "PASS",
        },
    }, ensure_ascii=False, indent=2))
    return result


V3_C248 = run_v3_c248()
````

````python
# NB22E-C250 — Descriptive paired policy comparison from finalized reviews.
# No generation, new judgment, ranking, Bootstrap, or model-alias acceptance.

def run_v3_c250():
    root = Path(
        "/Users/macbookpro/AMIR_DATA/00_WORKS/01_PROJECTS/"
        "RAG-projects/RAG_finance"
    ).resolve()
    folder = root / "reports/phase2/quality_improvement/NB22E.lab.v3"

    def require(ok, message):
        if not ok:
            raise ValueError(message)

    def sha(raw):
        return hashlib.sha256(raw).hexdigest()

    def enc(value):
        return (
            json.dumps(value, ensure_ascii=False, sort_keys=True, indent=2)
            + "\n"
        ).encode("utf-8")

    parent_pin = {
        "path": str((folder / (
            "C249.R11_common_rubric_final_dispositions."
            "7e0603d38b5aa16b6d308222a4f3063d84f61ed9df2ec5afe67116d4a427cc74.json"
        )).relative_to(root)),
        "file_bytes_sha256":
            "7e0603d38b5aa16b6d308222a4f3063d84f61ed9df2ec5afe67116d4a427cc74",
    }
    parent_path = root / parent_pin["path"]
    require(parent_path.is_file(), "C250_PARENT_MISSING")
    parent_raw = parent_path.read_bytes()
    require(
        sha(parent_raw) == parent_pin["file_bytes_sha256"],
        "C250_PARENT_HASH_MISMATCH",
    )
    parent = json.loads(parent_raw.decode("utf-8"))
    require(
        parent["unclassified_common_rubric_rows"] == 0,
        "C250_UNCLASSIFIED_ROWS_REMAIN",
    )

    pass_statuses = {
        "PASS_APPROVED_CONTENT_RUBRIC_MODEL_IDENTITY_HELD",
        "PASS_CONTENT",
        "PASS_CONTENT_WITH_FINDING",
    }
    fail_statuses = {
        "FAIL_FROZEN_ACTION_CRITERION_BARE_ABSTENTION",
        "FAIL_CONTENT",
        "FAIL_APPROVED_CONTENT_RUBRIC_MODEL_IDENTITY_HELD",
        "FAIL_CONTENT_CRITERION_ESTABLISHED",
        "FAIL_FROZEN_AMBIGUITY_NO_ABSTENTION_CRITERION",
    }
    nonbinary_statuses = {
        "UNDETERMINED_CONTENT",
        "UNDETERMINED_CURRENT_LINK",
        "SCOPED_LINK_SUPPORT_ONLY",
        "NOT_APPLICABLE_NOT_PASS",
        "NOT_APPLICABLE_NOT_A_CITATION_PASS",
    }

    def content_class(status):
        if status in pass_statuses:
            return "PASS"
        if status in fail_statuses:
            return "FAIL"
        require(status in nonbinary_statuses,
                "C250_UNRECOGNIZED_STATUS:" + status)
        return status

    grouped = {}
    for row in parent["paired_rows"]:
        spec_id = row["regression_spec_id"]
        arm = row["arm"]
        require(arm in {"reference", "candidate"}, "C250_UNKNOWN_ARM")
        require(arm not in grouped.setdefault(spec_id, {}),
                "C250_DUPLICATE_ARM_SCENARIO")
        grouped[spec_id][arm] = row

    require(len(grouped) == 28, "C250_EXPECTED_28_SCENARIOS")

    paired = []
    transitions = Counter()
    model_held_ids = []
    nonbinary_ids = []
    comparable_ids = []

    for spec_id, arms in sorted(grouped.items()):
        require(set(arms) == {"reference", "candidate"},
                "C250_MISSING_ARM:" + spec_id)
        reference = arms["reference"]
        candidate = arms["candidate"]
        require(
            reference["frozen_criteria"] == candidate["frozen_criteria"]
            and reference["expected_action"] == candidate["expected_action"],
            "C250_COMMON_CRITERIA_MISMATCH:" + spec_id,
        )

        old_class = content_class(reference["content_status"])
        new_class = content_class(candidate["content_status"])
        identity_held = any(
            row["model_status"] == "HOLD_NO_PROVIDER_MAPPING"
            for row in arms.values()
        )
        binary = (
            old_class in {"PASS", "FAIL"}
            and new_class in {"PASS", "FAIL"}
        )

        reasons = []
        if identity_held:
            model_held_ids.append(spec_id)
            reasons.append("MODEL_IDENTITY_NOT_QUALIFIED")
        if not binary:
            nonbinary_ids.append(spec_id)
            reasons.append("NONBINARY_OR_LINK_ONLY_DISPOSITION")

        transition = None
        if not reasons:
            transition = old_class + "_TO_" + new_class
            transitions[transition] += 1
            comparable_ids.append(spec_id)

        paired.append({
            "regression_spec_id": spec_id,
            "expected_action": reference["expected_action"],
            "reference_content_status": reference["content_status"],
            "candidate_content_status": candidate["content_status"],
            "reference_content_class": old_class,
            "candidate_content_class": new_class,
            "reference_answer_sha256": reference["answer_text_sha256"],
            "candidate_answer_sha256": candidate["answer_text_sha256"],
            "reference_model_status": reference["model_status"],
            "candidate_model_status": candidate["model_status"],
            "model_identity_hold": identity_held,
            "limited_descriptive_transition": transition,
            "comparison_hold_reasons": reasons,
            "name_consistency_is_full_runtime_equivalence": False,
            "formal_comparison_accepted": False,
        })

    require(
        sum(transitions.values()) == len(comparable_ids),
        "C250_TRANSITION_ACCOUNTING_MISMATCH",
    )

    result = {
        "cell": "C250",
        "status": "LIMITED_PAIRED_COMPARISON_REPORTED_WITH_EXPLICIT_HOLDS",
        "parent_review_receipt": parent_pin,
        "frozen_rubric_receipt": parent["frozen_rubric_receipt"],
        "scope": "28_APPROVED_PREVIOUSLY_SEEN_POLICY_SCENARIOS",
        "total_scenarios_accounted": len(paired),
        "arm_scenario_rows_accounted": 56,
        "limited_binary_pairs_without_model_identity_hold": len(comparable_ids),
        "limited_descriptive_transition_counts": dict(transitions),
        "limited_descriptive_scenario_ids": comparable_ids,
        "model_identity_held_scenario_count": len(model_held_ids),
        "model_identity_held_scenario_ids": model_held_ids,
        "nonbinary_or_link_only_scenario_count": len(nonbinary_ids),
        "nonbinary_or_link_only_scenario_ids": nonbinary_ids,
        "hold_categories_can_overlap": True,
        "paired_results": paired,
        "full_population_accuracy_computed": False,
        "unknown_or_NA_imputed_as_zero": False,
        "link_only_support_promoted_to_response_pass": False,
        "selection_based_on_winner_or_score": False,
        "58_pool50_groups_used_for_this_population": False,
        "model_alias_equivalence_granted": False,
        "provider_evidence_still_required": (
            "MAPPING_FOR_DEEPSEEK_SLASH_DEEPSEEK_V4_FLASH_ON_THE_USED_GATEWAY"
        ),
        "formal_nonregression_status": "NOT_ESTABLISHED_NOT_PASS",
        "historically_accepted_baseline_claimed": False,
        "independent_test_claimed": False,
        "product_acceptance_granted": False,
        "new_api_calls": 0,
        "new_semantic_judgments": 0,
        "new_ranking_runs": 0,
        "bootstrap_runs": 0,
        "prompt_changed": False,
        "historical_results_overwritten": False,
        "selected_version_gate": "BLOCKED",
    }

    raw = enc(result)
    digest = sha(raw)
    path = folder / f"C250.R11_limited_paired_policy_comparison.{digest}.json"
    if path.exists():
        require(path.read_bytes() == raw, "C250_OUTPUT_COLLISION")
    else:
        with path.open("xb") as handle:
            handle.write(raw)
            handle.flush()
            os.fsync(handle.fileno())

    require(sha(path.read_bytes()) == digest, "C250_RELOAD_FAILED")
    print(json.dumps({
        key: value for key, value in result.items() if key != "paired_results"
    } | {
        "decision_table": paired,
        "output": {
            "path": str(path.relative_to(root)),
            "file_bytes_sha256": digest,
            "reload_validation": "PASS",
        },
    }, ensure_ascii=False, indent=2))
    return result


V3_C250 = run_v3_c250()
````

````python
def run_v3_c253():
    root = Path(
        "/Users/macbookpro/AMIR_DATA/00_WORKS/01_PROJECTS/"
        "RAG-projects/RAG_finance"
    ).resolve()
    folder = root / "reports/phase2/quality_improvement/NB22E.lab.v3"
    inputs = {}

    def require(ok, message):
        if not ok:
            raise ValueError(message)

    def sha(raw):
        return hashlib.sha256(raw).hexdigest()

    def enc(value):
        return (
            json.dumps(value, ensure_ascii=False, sort_keys=True, indent=2)
            + "\n"
        ).encode("utf-8")

    def read_report(stem, digest):
        relative = (
            "reports/phase2/quality_improvement/NB22E.lab.v3/"
            + stem + "." + digest + ".json"
        )
        path = (root / relative).resolve()
        require(path.is_relative_to(root), "C253_PATH_OUTSIDE_PROJECT")
        require(path.is_file(), "C253_MISSING_EXACT_FILE:" + relative)
        raw = path.read_bytes()
        require(sha(raw) == digest, "C253_HASH_MISMATCH:" + relative)
        inputs[relative] = digest
        return json.loads(raw)

    def save(stem, value, suffix="json"):
        raw = enc(value) if suffix == "json" else value.encode("utf-8")
        digest = sha(raw)
        path = folder / (stem + "." + digest + "." + suffix)
        if path.exists():
            require(path.read_bytes() == raw, "C253_OUTPUT_CONFLICT")
        else:
            path.write_bytes(raw)
        require(sha(path.read_bytes()) == digest, "C253_RELOAD_HASH_MISMATCH")
        return {
            "path": path.relative_to(root).as_posix(),
            "file_bytes_sha256": digest,
            "reload_validation": "PASS",
        }

    lineage = read_report(
        "C252.R11_fixed_pool_reference_lineage",
        "ad1b918ddc64498a136d2e24028a13179d6e9508a09a2029ee00b7ad11a4d7cc",
    )
    initial = read_report(
        "C207.R11_independent_fixed_pool_arm_results",
        "e553f9f083ae6820eccf06cc2c69cdb017122637d38c64692448f11fade8726a",
    )
    completed = read_report(
        "C209.R11_fixed_pool_local_completion",
        "9cd518b16952474a2988ce7bdab4d291ec6b1aa98d843a14399d74e45cf52860",
    )

    require(
        lineage["lineage_compatible_query_count"] == 70
        and lineage["lineage_held_query_count"] == 0,
        "C253_LINEAGE_NOT_QUALIFIED",
    )
    metrics = ("hit_at_10", "recall_at_10", "ndcg_at_10")

    # Consume only records carrying the exact paired-metric schema.
    # Repeated identical records are provenance duplicates, not extra queries.
    values = {}
    value_origins = {}

    def collect(value, origin, pointer=""):
        if isinstance(value, dict):
            if (
                isinstance(value.get("candidate_id"), str)
                and isinstance(value.get("baseline_metrics"), dict)
                and isinstance(value.get("candidate_metrics"), dict)
                and all(
                    name in value["baseline_metrics"]
                    and name in value["candidate_metrics"]
                    for name in metrics
                )
            ):
                candidate_id = value["candidate_id"]
                pair = {
                    arm: {
                        name: value[arm + "_metrics"][name]
                        for name in metrics
                    }
                    for arm in ("baseline", "candidate")
                }
                for arm in pair.values():
                    for number in arm.values():
                        require(
                            isinstance(number, (int, float))
                            and not isinstance(number, bool)
                            and 0 <= number <= 1,
                            "C253_INVALID_METRIC:" + candidate_id,
                        )
                if candidate_id in values:
                    require(
                        values[candidate_id] == pair,
                        "C253_CONFLICTING_PAIRED_METRICS:" + candidate_id,
                    )
                else:
                    values[candidate_id] = pair
                value_origins.setdefault(candidate_id, []).append({
                    "source": origin,
                    "json_pointer": pointer,
                })
            for key, child in value.items():
                collect(child, origin, pointer + "/" + str(key))
        elif isinstance(value, list):
            for number, child in enumerate(value):
                collect(child, origin, pointer + "/" + str(number))

    collect(initial, "C207")
    collect(completed, "C209")

    membership = {}
    for row in lineage["query_reviews"]:
        require(row["status"] == "LINEAGE_COMPATIBLE", "C253_QUERY_HELD")
        candidate_id = row["candidate_id"]
        require(candidate_id not in membership, "C253_DUPLICATE_QUERY")
        membership[candidate_id] = row["source_component"]

    require(len(membership) == 70, "C253_EXPECTED_70_QUERIES")
    require(
        set(membership).issubset(values),
        "C253_MISSING_PAIRED_METRICS:"
        + ",".join(sorted(set(membership) - set(values))),
    )
    values = {key: values[key] for key in sorted(membership)}
    groups = {}
    for candidate_id in sorted(membership):
        groups.setdefault(membership[candidate_id], []).append(candidate_id)
    require(len(groups) == 58, "C253_EXPECTED_58_PARENT_GROUPS")

    preserved_counts = completed["scenario_status_counts"]
    require(
        preserved_counts == {
            "SCOPED_REPAIR_TARGET_NOT_MET": 73,
            "SCOPED_REPAIR_TARGET_MET": 31,
        },
        "C253_SCENARIO_FINDINGS_CHANGED",
    )

    plan = {
        "cell": "C253",
        "arm_id": "pool50_char_rrf_v1",
        "authorization": "EXPLICIT_SUPERVISOR_APPROVAL_IN_CURRENT_CHAT",
        "input_file_hashes": dict(inputs),
        "method": "PAIRED_PARENT_GROUP_BOOTSTRAP",
        "resamples": 10000,
        "seed": 20261027,
        "confidence_level": 0.95,
        "interval": "TWO_SIDED_PERCENTILE_2_5_AND_97_5",
        "percentile_interpolation": "LINEAR_AT_P_TIMES_N_MINUS_ONE",
        "groups_sampled_with_replacement_per_replicate": 58,
        "sampling_order": "SORTED_PARENT_GROUP_IDS",
        "all_group_queries_retained_with_multiplicity": True,
        "same_resampled_queries_in_both_arms": True,
        "aggregation": "ARITHMETIC_MEAN_OF_QUERY_METRICS",
        "group_means_equally_weighted": False,
        "difference": "CANDIDATE_MINUS_BASELINE",
        "metrics": list(metrics),
        "simultaneous_interval_claimed": False,
        "formal_acceptance_rule_applied": False,
        "membership": membership,
        "paired_query_metrics": values,
        "metric_record_origins": {
            key: value_origins[key] for key in sorted(membership)
        },
    }
    plan_pin = save("C253.R11_fixed_pool_bootstrap_plan", plan)

    # Standalone stdlib runner: no Jupyter state or API access is needed.
    runner_source = r'''
import hashlib
import json
import random
import sys
from pathlib import Path

path = Path(sys.argv[1])
expected = sys.argv[2]
raw = path.read_bytes()
if hashlib.sha256(raw).hexdigest() != expected:
    raise ValueError("C253_CHILD_PLAN_HASH_MISMATCH")
plan = json.loads(raw)
metrics = plan["metrics"]
values = plan["paired_query_metrics"]
groups = {}
for candidate_id, group in sorted(plan["membership"].items()):
    groups.setdefault(group, []).append(candidate_id)
group_ids = sorted(groups)
if len(group_ids) != 58 or len(values) != 70:
    raise ValueError("C253_CHILD_POPULATION_MISMATCH")

# Group sums preserve query weighting and repeated group multiplicity.
stats = []
for group in group_ids:
    ids = groups[group]
    stats.append((
        len(ids),
        [sum(values[q]["baseline"][m] for q in ids) for m in metrics],
        [sum(values[q]["candidate"][m] for q in ids) for m in metrics],
    ))

rng = random.Random(plan["seed"])
samples = {m: [] for m in metrics}
for _ in range(plan["resamples"]):
    n = 0
    baseline = [0.0] * len(metrics)
    candidate = [0.0] * len(metrics)
    for _ in range(len(group_ids)):
        size, b, c = stats[rng.randrange(len(group_ids))]
        n += size
        for j in range(len(metrics)):
            baseline[j] += b[j]
            candidate[j] += c[j]
    for j, metric in enumerate(metrics):
        samples[metric].append((candidate[j] - baseline[j]) / n)

def percentile(numbers, probability):
    position = probability * (len(numbers) - 1)
    lower = int(position)
    upper = min(lower + 1, len(numbers) - 1)
    fraction = position - lower
    return numbers[lower] * (1 - fraction) + numbers[upper] * fraction

results = {}
for metric in metrics:
    baseline = sum(v["baseline"][metric] for v in values.values()) / 70
    candidate = sum(v["candidate"][metric] for v in values.values()) / 70
    ordered = sorted(samples[metric])
    low = percentile(ordered, 0.025)
    high = percentile(ordered, 0.975)
    results[metric] = {
        "baseline": baseline,
        "candidate": candidate,
        "candidate_minus_baseline": candidate - baseline,
        "difference_percentile_95_interval": [low, high],
        "individual_interval_lower_bound_above_zero": low > 0,
    }

print(json.dumps({
    "status": "PAIRED_GROUP_BOOTSTRAP_COMPLETED",
    "resamples": plan["resamples"],
    "seed": plan["seed"],
    "query_count": len(values),
    "parent_group_count": len(group_ids),
    "metrics": results,
    "python_version": sys.version,
    "rng": "PYTHON_STDLIB_RANDOM_MT19937",
}, sort_keys=True))
'''
    runner_pin = save(
        "C253.fixed_pool_group_bootstrap_runner", runner_source, suffix="py"
    )
    child = subprocess.run(
        [
            sys.executable,
            str(root / runner_pin["path"]),
            str(root / plan_pin["path"]),
            plan_pin["file_bytes_sha256"],
        ],
        capture_output=True,
        text=True,
        timeout=120,
        check=False,
    )
    if child.returncode != 0:
        failure = save("C253.R11_bootstrap_failure", {
            "plan": plan_pin,
            "runner": runner_pin,
            "returncode": child.returncode,
            "stdout": child.stdout[-4000:],
            "stderr": child.stderr[-4000:],
        })
        raise RuntimeError("C253_BOOTSTRAP_FAILED:" + failure["path"])

    calculation = json.loads(child.stdout)
    require(
        calculation["status"] == "PAIRED_GROUP_BOOTSTRAP_COMPLETED",
        "C253_CHILD_STATUS_INVALID",
    )

    # Verify point estimates against C209 without replacing its historical record.
    for metric in metrics:
        previous = completed["paired_descriptive_comparison"]["metrics"][metric]
        current = calculation["metrics"][metric]
        for key in ("baseline", "candidate", "candidate_minus_baseline"):
            require(
                abs(previous[key] - current[key]) < 1e-12,
                "C253_C209_POINT_ESTIMATE_MISMATCH:" + metric + ":" + key,
            )

    result = {
        "cell": "C253",
        "roadmap_subtask": "R11.3",
        "checklist_item": 10,
        "status": "FIXED_POOL_PAIRED_GROUP_INTERVALS_RECORDED",
        "arm_id": "pool50_char_rrf_v1",
        "plan_receipt": plan_pin,
        "runner_receipt": runner_pin,
        "input_file_hashes": inputs,
        "calculation": calculation,
        "scope": "INDEPENDENT_POOL50_ARM_AND_SAME_70_QUERIES_ONLY",
        "baseline_interpretation": (
            "SAVED_COMPARATOR_ORDER;"
            "NOT_A_NEW_HISTORICAL_ACCEPTANCE_CLAIM"
        ),
        "intervals_are_individual": True,
        "simultaneous_coverage_claimed": False,
        "formal_nonregression_granted": False,
        "formal_acceptance_disposition": (
            "INTERVALS_COMPUTED;NO_ADDITIONAL_ACCEPTANCE_RULE_APPLIED"
        ),
        "scenario_status_counts_preserved": preserved_counts,
        "all_scenario_repair_targets_met": False,
        "declared_pool_156_acceptance_granted": False,
        "selected_version_acceptance_granted": False,
        "C250_package_reopened": False,
        "model_identity_hold_preserved": True,
        "historical_results_overwritten": False,
        "done_item_count": 9,
        "open_item_count": 3,
        "open_item_ids": [9, 10, 12],
        "item10_complete": False,
        "new_api_calls": 0,
        "new_ranking_runs": 0,
        "bootstrap_runs": 1,
        "selected_version_gate": "BLOCKED",
    }
    output = save("C253.R11_fixed_pool_paired_group_intervals", result)
    print(json.dumps(
        {
            key: value for key, value in result.items()
            if key != "input_file_hashes"
        } | {"output": output},
        ensure_ascii=False,
        indent=2,
    ))
    return result

V3_C253 = run_v3_c253()
````

````python
def run_v3_c256():
    root = Path(
        "/Users/macbookpro/AMIR_DATA/00_WORKS/01_PROJECTS/"
        "RAG-projects/RAG_finance"
    ).resolve()
    relative = (
        "reports/phase2/quality_improvement/NB22E.lab.v3/"
        "C189.R11_remaining_selected_rank_inputs."
        "54b17688cd7f96885984b8f6a9df2f64b53bd636090128269ea12b54c0422f5f.json"
    )
    expected = (
        "54b17688cd7f96885984b8f6a9df2f64b53bd636090128269ea12b54c0422f5f"
    )
    path = (root / relative).resolve()
    if not path.is_relative_to(root):
        raise ValueError("C256_PATH_OUTSIDE_PROJECT")
    if not path.is_file():
        raise ValueError("C256_MISSING_EXACT_FILE:" + relative)

    raw = path.read_bytes()
    if hashlib.sha256(raw).hexdigest() != expected:
        raise ValueError("C256_C189_HASH_MISMATCH")
    source = json.loads(raw)

    bindings = []
    named_hashes = []
    producer_fields = []
    schemas = {}

    def walk(value, pointer=""):
        if isinstance(value, dict):
            schemas.setdefault(pointer or "/", sorted(value))
            bound_path = value.get("path")
            digest = value.get("file_bytes_sha256")
            if isinstance(bound_path, str) and isinstance(digest, str):
                bindings.append({
                    "json_pointer": pointer,
                    "path": bound_path,
                    "file_bytes_sha256": digest,
                })

            for key, child in value.items():
                location = pointer + "/" + str(key)
                name = str(key).lower()

                if isinstance(child, str) and re.fullmatch(
                    r"[0-9a-f]{64}", child
                ):
                    named_hashes.append({
                        "json_pointer": location,
                        "name": str(key),
                        "sha256": child,
                    })

                if (
                    any(token in name for token in (
                        "version", "producer", "implementation",
                        "configuration", "provider", "rank_input_mode",
                        "pipeline_source", "arm_id", "selected_variant"
                    ))
                    and isinstance(child, (str, int, float, bool))
                ):
                    producer_fields.append({
                        "json_pointer": location,
                        "recorded_value": child,
                    })

                if isinstance(child, list):
                    first_dict = next(
                        (item for item in child if isinstance(item, dict)),
                        None,
                    )
                    schemas[location] = {
                        "type": "list",
                        "count": len(child),
                        "first_record_keys": (
                            sorted(first_dict) if first_dict is not None else None
                        ),
                    }
                walk(child, location)

        elif isinstance(value, list):
            for number, child in enumerate(value):
                walk(child, pointer + "/" + str(number))

    walk(source)

    # Consolidate identical path/hash bindings without discarding their origins.
    by_binding = {}
    for row in bindings:
        key = (row["path"], row["file_bytes_sha256"])
        by_binding.setdefault(key, []).append(row["json_pointer"])

    connections = [{
        "path": key[0],
        "file_bytes_sha256": key[1],
        "recorded_at": pointers,
    } for key, pointers in sorted(by_binding.items())]

    result = {
        "cell": "C256",
        "status": "SELECTED_RANK_PRODUCER_CONNECTIONS_EXTRACTED_READ_ONLY",
        "source_receipt": {
            "path": relative,
            "file_bytes_sha256": expected,
        },
        "top_level_keys": sorted(source),
        "producer_fields": producer_fields,
        "named_hash_bindings": [
            row for row in named_hashes
            if row["json_pointer"].count("/") <= 3
        ],
        "direct_connections": connections,
        "top_level_collection_schemas": {
            pointer: schema
            for pointer, schema in schemas.items()
            if pointer.count("/") <= 1
        },
        "source_declared_candidate_count": source.get("candidate_count"),
        "source_declared_completed_candidate_count": source.get(
            "completed_candidate_count"
        ),
        "pool_completeness_granted": False,
        "selected_version_equivalence_granted": False,
        "new_execution_required": "NOT_YET_DETERMINED",
        "files_written": 0,
        "new_api_calls": 0,
        "new_ranking_runs": 0,
        "bootstrap_runs": 0,
        "C250_package_reopened": False,
        "C253_aggregate_comparison_preserved": True,
        "selected_version_gate": "BLOCKED",
    }
    print(json.dumps(result, ensure_ascii=False, indent=2))
    return result

V3_C256 = run_v3_c256()
````

### C257 — مصرف ترتیب‌های کامل موجود C189 برای کنترل Pool

رسیدهای C189 با هش ثبت‌شده خوانده می‌شوند. یکتایی شناسه‌ها، اجتماع دو ورودی رتبه‌بندی، اتصال قرارداد و هش سؤال کنترل می‌شوند؛ سپس حضور چانک‌های مرجع در همان مجموعه بررسی می‌شود.

موارد فاقد رسید یا دارای اختلاف نسخه جدا باقی می‌مانند. این کنترل، محاسبهٔ مجدد رتبه‌ها نیست و ضرورت اجرای تازه را برای موارد پوشش‌داده‌نشده اثبات نمی‌کند.

````python
def run_v3_c257():
    root = Path(
        "/Users/macbookpro/AMIR_DATA/00_WORKS/01_PROJECTS/"
        "RAG-projects/RAG_finance"
    ).resolve()
    folder = root / "reports/phase2/quality_improvement/NB22E.lab.v3"
    inputs = {}

    def require(ok, message):
        if not ok:
            raise ValueError(message)

    def sha(raw):
        return hashlib.sha256(raw).hexdigest()

    def canonical(value):
        return json.dumps(
            value, ensure_ascii=False, sort_keys=True,
            separators=(",", ":"), allow_nan=False,
        ).encode("utf-8")

    def read_pin(relative, digest, jsonl=False):
        path = (root / relative).resolve()
        require(path.is_relative_to(root), "C257_PATH_OUTSIDE_PROJECT")
        require(path.is_file(), "C257_MISSING_EXACT_FILE:" + relative)
        raw = path.read_bytes()
        require(sha(raw) == digest, "C257_HASH_MISMATCH:" + relative)
        inputs[relative] = digest
        if jsonl:
            return [
                json.loads(line)
                for line in raw.decode("utf-8").splitlines()
                if line.strip()
            ]
        return json.loads(raw)

    def report(stem, digest):
        return read_pin(
            "reports/phase2/quality_improvement/NB22E.lab.v3/"
            + stem + "." + digest + ".json", digest,
        )

    pending = report(
        "C255.R11_declared_pool_missing_receipt_requirements",
        "cb9ea2298c273b055ccebe4e1f77ef880a05c989b27f674a163d9846bf81fb30",
    )
    ranks = report(
        "C189.R11_remaining_selected_rank_inputs",
        "54b17688cd7f96885984b8f6a9df2f64b53bd636090128269ea12b54c0422f5f",
    )
    golden = read_pin(
        "artifacts/phase2/golden_dataset_v2/golden_dataset_v2.jsonl",
        "b9ca18e710686ddca48a25c637840b716adb306717cafd8ec7c96ce9e182e6f2",
        jsonl=True,
    )
    require(len(pending["scenario_results"]) == 156, "C257_EXPECTED_156")
    contract_sha = sha(canonical(ranks["contract"]))
    require(
        contract_sha == ranks["contract_sha256"],
        "C257_CONTRACT_HASH_MISMATCH",
    )

    golden_by_id = {}
    for row in golden:
        candidate_id = row["candidate_id"]
        require(candidate_id not in golden_by_id, "C257_DUPLICATE_GOLDEN_ID")
        golden_by_id[candidate_id] = row

    receipts = ranks["candidate_receipts"]
    require(len(receipts) == 33, "C257_EXPECTED_33_BOUND_RECEIPTS")
    checkpoints = {}
    for pin in receipts:
        candidate_id = pin["candidate_id"]
        require(candidate_id not in checkpoints, "C257_DUPLICATE_CHECKPOINT")
        saved = read_pin(pin["path"], pin["file_bytes_sha256"])
        require(
            saved["candidate_id"] == candidate_id,
            "C257_RECEIPT_CANDIDATE_MISMATCH",
        )
        checkpoints[candidate_id] = (pin, saved)

    def valid_order(value):
        return (
            isinstance(value, list)
            and all(isinstance(x, str) for x in value)
            and len(value) == len(set(value))
        )

    results = []
    for target in sorted(
        pending["scenario_results"], key=lambda x: x["regression_spec_id"]
    ):
        candidate_id = target["candidate_id"]
        base = {
            "regression_spec_id": target["regression_spec_id"],
            "candidate_id": candidate_id,
            "formal_scenario_pass": False,
        }
        if candidate_id not in checkpoints:
            results.append(base | {
                "status": "NO_MATCHING_CHECKPOINT_IN_C189_PACKET",
                "required_evidence": (
                    "COMPATIBLE_COMPLETE_POOL_RECEIPT_FOR_THIS_QUERY"
                ),
            })
            continue

        pin, saved = checkpoints[candidate_id]
        require(candidate_id in golden_by_id, "C257_GOLDEN_QUERY_MISSING")
        question = golden_by_id[candidate_id]["evaluation_content"]["question"]
        issues = []

        if saved["contract_sha256"] != contract_sha:
            issues.append("CONTRACT_MISMATCH")
        if saved["question_sha256"] != sha(question.encode("utf-8")):
            issues.append("QUESTION_VERSION_HASH_MISMATCH")
        if saved["ranker_question_sha256"] != sha(
            question.strip().encode("utf-8")
        ):
            issues.append("RANKER_QUESTION_HASH_MISMATCH")

        fields = (
            "hybrid100_order", "local100_order", "quota100_order",
            "char_channel_order", "char_rrf_order", "char_rrf_top10",
        )
        for field in fields:
            if not valid_order(saved[field]):
                issues.append("INVALID_ORDER:" + field)

        if not issues:
            hybrid = saved["hybrid100_order"]
            local = saved["local100_order"]
            quota = saved["quota100_order"]
            channel = saved["char_channel_order"]
            fused = saved["char_rrf_order"]

            if not (
                len(hybrid) == len(local) == len(quota) == 100
                and set(hybrid) == set(local) == set(quota)
            ):
                issues.append("INCUMBENT100_MEMBERSHIP_CONFLICT")
            if set(fused) != set(quota) | set(channel):
                issues.append("COMPLETE_FUSION_MEMBERSHIP_CONFLICT")
            if saved["char_rrf_top10"] != fused[:10]:
                issues.append("TOP10_PREFIX_CONFLICT")
            if len(fused) != pin["char_rrf_depth"]:
                issues.append("DECLARED_FUSION_DEPTH_CONFLICT")
            if len(channel) != pin["char_channel_depth"]:
                issues.append("DECLARED_CHAR_DEPTH_CONFLICT")

        if issues:
            results.append(base | {
                "status": "CHECKPOINT_COMPATIBILITY_HOLD",
                "receipt": pin,
                "issues": issues,
            })
            continue

        references = set(target["historical_reference_chunk_ids"])
        require(bool(references), "C257_EMPTY_REFERENCE_MAPPING")
        pool = set(saved["char_rrf_order"])
        present = sorted(references & pool)
        results.append(base | {
            "status": (
                "REFERENCE_PRESENT_IN_COMPLETE_C189_ORDER"
                if present else "REFERENCE_ABSENT_FROM_COMPLETE_C189_ORDER"
            ),
            "receipt": pin,
            "contract_sha256": contract_sha,
            "complete_order_count": len(pool),
            "reference_chunk_ids": sorted(references),
            "reference_ids_present": present,
            "reference_ids_absent": sorted(references - pool),
            "complete_membership_union_control": "PASS",
            "evidence_scope": "THIS_VERSION_BOUND_C189_EXECUTION",
            "general_selected_runtime_equivalence_granted": False,
        })

    counts = {}
    for row in results:
        counts[row["status"]] = counts.get(row["status"], 0) + 1

    result = {
        "cell": "C257",
        "roadmap_subtask": "R11.3",
        "checklist_item": 10,
        "status": "EXISTING_C189_COMPLETE_ORDER_MEMBERSHIP_CONTROLLED",
        "input_file_hashes": inputs,
        "contract": ranks["contract"],
        "contract_sha256": contract_sha,
        "scenario_count": 156,
        "available_C189_checkpoint_count": len(checkpoints),
        "target_queries_with_C189_checkpoint": sum(
            row["candidate_id"] in checkpoints for row in results
        ),
        "scenario_status_counts": counts,
        "scenario_results": results,
        "membership_rule": "SET_CHAR_RRF_EQUALS_SET_QUOTA100_UNION_CHAR_CHANNEL",
        "ranking_scores_recomputed": False,
        "historical_C036_relabelled_as_selected_execution": False,
        "new_formal_scenario_passes": 0,
        "general_selected_runtime_acceptance_granted": False,
        "repository_wide_cache_absence_claimed": False,
        "definite_new_execution_count": None,
        "remaining_requirement": (
            "QUALIFY_REMAINING_EXACT_QUERY_RECEIPTS_AND_SELECTED_PRODUCER;"
            "DO_NOT_ASSUME_EVERY_UNCOVERED_QUERY_REQUIRES_NEW_EXECUTION"
        ),
        "C250_package_reopened": False,
        "C253_aggregate_comparison_preserved": True,
        "historical_results_overwritten": False,
        "new_api_calls": 0,
        "new_model_runs": 0,
        "new_retrieval_runs": 0,
        "new_ranking_runs": 0,
        "bootstrap_runs": 0,
        "item10_complete": False,
        "selected_version_gate": "BLOCKED",
    }
    raw = (
        json.dumps(result, ensure_ascii=False, sort_keys=True, indent=2)
        + "\n"
    ).encode("utf-8")
    digest = sha(raw)
    output = folder / (
        "C257.R11_existing_complete_pool_membership." + digest + ".json"
    )
    if output.exists():
        require(output.read_bytes() == raw, "C257_OUTPUT_CONFLICT")
    else:
        output.write_bytes(raw)
    require(sha(output.read_bytes()) == digest, "C257_RELOAD_HASH_MISMATCH")
    require(json.loads(output.read_bytes()) == result, "C257_RELOAD_MISMATCH")

    print(json.dumps(
        {
            key: value for key, value in result.items()
            if key not in ("scenario_results", "input_file_hashes")
        } | {
            "compatibility_hold_examples": [
                row for row in results
                if row["status"] == "CHECKPOINT_COMPATIBILITY_HOLD"
            ][:3],
            "output": {
                "path": output.relative_to(root).as_posix(),
                "file_bytes_sha256": digest,
                "reload_validation": "PASS",
            },
        },
        ensure_ascii=False,
        indent=2,
    ))
    return result

V3_C257 = run_v3_c257()
````

نیازمندی‌های تاریخی ثبت‌شده با پوشش واحدبه‌واحد پاسخ فعلی یکی نیستند. هیچ داوری، تولید پاسخ یا سنجهٔ کیفیت تازه‌ای اجرا نمی‌شود.

````python
# NB22E v3 | C266 | R11.2
# Action: Bind existing scoped reviews to primary-v1 responses and identify metric-unit gaps.

def run_v3_c266():
    root = Path(
        "/Users/macbookpro/AMIR_DATA/00_WORKS/01_PROJECTS/"
        "RAG-projects/RAG_finance"
    ).resolve()
    folder = root / "reports/phase2/quality_improvement/NB22E.lab.v3"
    inputs = {}

    def require(ok, message):
        if not ok:
            raise ValueError(message)

    def sha(raw):
        return hashlib.sha256(raw).hexdigest()

    def read_pin(pin):
        path = (root / pin["path"]).resolve()
        require(path.is_relative_to(root), "C266_PATH_OUTSIDE_PROJECT")
        require(path.is_file(), "C266_MISSING_EXACT_FILE:" + pin["path"])
        raw = path.read_bytes()
        require(
            sha(raw) == pin["file_bytes_sha256"],
            "C266_HASH_MISMATCH:" + pin["path"],
        )
        inputs[pin["path"]] = pin["file_bytes_sha256"]
        return json.loads(raw)

    manifest = read_pin({
        "path": (
            "reports/phase2/quality_improvement/NB22E.lab.v3/"
            "C265.metric_binding_input_manifest."
            "788edcd46daffe93f6d6a146e05edc6d06f3fc5a1eb39c232ef2dc3a04338436.json"
        ),
        "file_bytes_sha256": (
            "788edcd46daffe93f6d6a146e05edc6d06f3fc5a1eb39c232ef2dc3a04338436"
        ),
    })
    documents = {
        pin["role"]: read_pin(pin) for pin in manifest["files"]
    }
    primary = documents["PRIMARY_V1_RESPONSE_VERSION_BINDINGS"]
    historical = documents["EXISTING_HISTORICAL_REFERENCE_EVALUATION_REPORT"]
    table = documents["EXISTING_HISTORICAL_METRIC_TABLE"]
    rubric = documents[
        "APPROVED_LIMITED_RUBRIC_NOT_AUTOMATIC_FORMAL_METRIC_REFERENCE"
    ]
    reviews = documents["EXISTING_SCOPED_RESPONSE_DISPOSITIONS_NO_REJUDGING"]

    rubric_by_spec = {}
    for row in rubric["rows"]:
        key = row["regression_spec_id"]
        require(key not in rubric_by_spec, "C266_DUPLICATE_RUBRIC_SPEC")
        rubric_by_spec[key] = row
    require(len(rubric_by_spec) == 35, "C266_EXPECTED_35_RUBRIC_ROWS")

    reference_reviews = {}
    for row in reviews["paired_rows"]:
        if row["arm"] != "reference":
            continue
        key = (row["candidate_id"], row["regression_spec_id"])
        require(key not in reference_reviews, "C266_DUPLICATE_REFERENCE_REVIEW")
        reference_reviews[key] = row

    historical_by_id = {}
    for row in historical["case_measurements"]:
        candidate_id = row["candidate_id"]
        require(
            candidate_id not in historical_by_id,
            "C266_DUPLICATE_HISTORICAL_CASE",
        )
        historical_by_id[candidate_id] = row

    completeness_rows = [
        row for row in table if row["metric"] == "answer_completeness"
    ]
    require(len(completeness_rows) == 1, "C266_COMPLETENESS_ROW_AMBIGUOUS")
    completeness = completeness_rows[0]
    require(
        completeness["status"]
        == "RECORDED_REQUIREMENTS_AVAILABLE_NO_PER_REQUIREMENT_COVERAGE",
        "C266_HISTORICAL_COMPLETENESS_EVIDENCE_CHANGED",
    )

    rows = []
    seen = set()
    for answer in primary["candidate_evidence_rows"]:
        candidate_id = answer["candidate_id"]
        for spec_id in answer["scenario_ids"]:
            require(spec_id not in seen, "C266_DUPLICATE_SCENARIO")
            seen.add(spec_id)
            rule = rubric_by_spec[spec_id]
            require(
                rule["candidate_id"] == candidate_id,
                "C266_RUBRIC_CANDIDATE_MISMATCH",
            )
            prior = reference_reviews.get((candidate_id, spec_id))
            context_hash = rule["required_evidence"].get("context_sha256")
            answer_matches = (
                prior is not None
                and isinstance(answer.get("answer_text_sha256"), str)
                and prior["answer_text_sha256"] == answer["answer_text_sha256"]
            )
            context_matches = (
                context_hash is not None
                and context_hash == answer["rendered_context_sha256"]
            )
            exact_model = (
                answer.get("model_identity_status") == "EXACT_IDENTIFIER_MATCH"
            )

            if rule["rubric_status"] != "APPROVED_FOR_LIMITED_POLICY_COMPARISON":
                disposition = "HOLD_REFERENCE_OR_EXPECTATION"
            elif not answer_matches:
                disposition = "DO_NOT_TRANSFER_REVIEW_FROM_DIFFERENT_RESPONSE"
            elif not exact_model:
                disposition = "HOLD_MODEL_IDENTITY"
            elif context_hash is None:
                disposition = "CURRENT_LINK_EVIDENCE_BOUNDARY_REMAINS_SEPARATE"
            elif not context_matches:
                disposition = "HOLD_CONTEXT_VERSION_MISMATCH"
            else:
                disposition = "REUSE_EXISTING_LIMITED_CONTENT_REVIEW"

            old_case = historical_by_id.get(candidate_id)
            rows.append({
                "candidate_id": candidate_id,
                "regression_spec_id": spec_id,
                "primary_response_binding_status": answer["status"],
                "primary_answer_text_sha256": answer.get("answer_text_sha256"),
                "existing_review_answer_text_sha256": (
                    prior["answer_text_sha256"] if prior else None
                ),
                "answer_hash_matches": answer_matches,
                "rubric_context_sha256": context_hash,
                "primary_context_sha256": answer["rendered_context_sha256"],
                "context_hash_matches": context_matches,
                "model_identity_status": answer.get("model_identity_status"),
                "disposition": disposition,
                "existing_content_status_preserved": (
                    prior["content_status"] if prior else None
                ),
                "existing_content_review_receipt": (
                    prior.get("content_review_receipt") if prior else None
                ),
                "existing_response_receipts": (
                    prior["response_receipts"] if prior else []
                ),
                "rubric_status": rule["rubric_status"],
                "rubric_official_reference_approved": rule[
                    "official_reference_approved"
                ],
                "existing_criteria_preserved": rule["common_acceptance_criteria"],
                "historical_case_present": old_case is not None,
                "historical_case_contains_response_hash": (
                    old_case is not None and any(
                        key in old_case for key in (
                            "answer_text_sha256", "response_document_sha256",
                            "response_text_sha256",
                        )
                    )
                ),
                "historical_decision_transferred_to_current_answer": False,
                "formal_completeness_units_qualified_by_this_packet": False,
                "formal_completeness_score": None,
                "new_judgment_created": False,
            })

    require(seen == set(rubric_by_spec), "C266_SCENARIO_COVERAGE_MISMATCH")
    counts = {}
    for row in rows:
        key = row["disposition"]
        counts[key] = counts.get(key, 0) + 1

    result = {
        "cell": "C266",
        "roadmap_subtask": "R11.2",
        "status": "EXISTING_REVIEW_REUSE_AND_METRIC_UNIT_GAPS_RESOLVED_IN_PACKET",
        "input_file_hashes": inputs,
        "primary_target_identity_sha256": primary[
            "primary_target_identity_sha256"
        ],
        "scenario_row_count": len(rows),
        "disposition_counts": counts,
        "counts_are_unique_requests": False,
        "scenario_evidence_ledger": rows,
        "historical_completeness_witness": {
            key: completeness.get(key) for key in (
                "status", "cases_with_recorded_requirements",
                "recorded_requirement_count",
                "explicit_validated_result_count", "reason",
            )
        },
        "completeness_gap_decision": {
            "status": "UNIT_LEVEL_COVERAGE_AND_REFERENCE_QUALIFICATION_NOT_ESTABLISHED",
            "existing_decision_reviews_reused_without_rejudging": True,
            "missing_evidence": [
                "QUALIFIED_REQUIRED_INFORMATION_UNITS_FOR_EACH_ELIGIBLE_QUERY",
                "UNIT_SPECIFIC_COVERAGE_VERDICTS_BOUND_TO_EXACT_PRIMARY_ANSWER",
                "FROZEN_ELIGIBILITY_AND_COMPLETE_DENOMINATOR",
            ],
            "recorded_historical_requirement_counts_are_current_answer_coverage": False,
            "limited_policy_rubric_promoted_to_formal_metric_reference": False,
            "historical_report_candidate_match_is_version_equivalence": False,
            "missing_evidence_is_zero_or_model_failure": False,
            "new_generation_or_judging_proposed": False,
        },
        "citation_boundary": (
            "MATCHED_RESPONSE_TEXT_DOES_NOT_REPLACE_CURRENT_CLAIM_LINK_"
            "AND_LOCATOR_ASSESSMENT;N_A_IS_NOT_A_CITATION_PASS"
        ),
        "historical_results_overwritten": False,
        "C250_package_reopened": False,
        "C260_item10_disposition_preserved": True,
        "candidate_v2_1_results_merged": False,
        "new_api_calls": 0,
        "new_model_runs": 0,
        "new_semantic_judgments": 0,
        "new_metric_runs": 0,
        "formal_metric_gaps_closed": 0,
        "formal_nonregression_status": "NOT_COMPARABLE_NOT_PASS",
        "R11_2_complete": False,
        "selected_version_gate": "BLOCKED",
    }
    raw = (
        json.dumps(result, ensure_ascii=False, sort_keys=True, indent=2) + "\n"
    ).encode("utf-8")
    digest = sha(raw)
    output = folder / (
        "C266.R11_existing_review_and_metric_unit_bindings." + digest + ".json"
    )
    if output.exists():
        require(output.read_bytes() == raw, "C266_OUTPUT_CONFLICT")
    else:
        with output.open("xb") as stream:
            stream.write(raw)
    require(sha(output.read_bytes()) == digest, "C266_RELOAD_HASH_MISMATCH")
    require(json.loads(output.read_bytes()) == result, "C266_RELOAD_MISMATCH")

    print(json.dumps(
        {
            key: value for key, value in result.items()
            if key not in ("scenario_evidence_ledger", "input_file_hashes")
        } | {
            "output": {
                "path": output.relative_to(root).as_posix(),
                "file_bytes_sha256": digest,
                "reload_validation": "PASS",
            }
        },
        ensure_ascii=False, indent=2,
    ))
    return result

V3_C266 = run_v3_c266()
````

### C267 — اتصال نیازمندی‌های تاریخی به زمینهٔ فعلی نسخهٔ اول

````python
# NB22E v3 | C267 | R11.2
# Action: Bind recorded requirement units and check historical witness availability in current contexts.

def run_v3_c267():
    root = Path(
        "/Users/macbookpro/AMIR_DATA/00_WORKS/01_PROJECTS/"
        "RAG-projects/RAG_finance"
    ).resolve()
    folder = root / "reports/phase2/quality_improvement/NB22E.lab.v3"
    inputs = {}

    def require(ok, message):
        if not ok:
            raise ValueError(message)

    def sha(raw):
        return hashlib.sha256(raw).hexdigest()

    def canonical(value):
        return json.dumps(
            value, ensure_ascii=False, sort_keys=True,
            separators=(",", ":"), allow_nan=False,
        ).encode("utf-8")

    def read_raw(relative, digest):
        path = (root / relative).resolve()
        require(path.is_relative_to(root), "C267_PATH_OUTSIDE_PROJECT")
        require(path.is_file(), "C267_MISSING_EXACT_FILE:" + relative)
        raw = path.read_bytes()
        require(sha(raw) == digest, "C267_HASH_MISMATCH:" + relative)
        inputs[relative] = digest
        return raw

    def read_pin(pin):
        return json.loads(read_raw(pin["path"], pin["file_bytes_sha256"]))

    producer_path = (
        "artifacts/phase2/evaluation_and_ablation/work/outputs/"
        "NB21C.C017.corrective_semantic219_v1/"
        "C017.report."
        "b690f2260ab1dd686411dfe56a8679642b9f3f906e22c8eaa95e815517ff737c.json"
    )
    producer_hash = (
        "b690f2260ab1dd686411dfe56a8679642b9f3f906e22c8eaa95e815517ff737c"
    )
    producer = json.loads(read_raw(producer_path, producer_hash))
    primary = read_pin({
        "path": (
            "reports/phase2/quality_improvement/NB22E.lab.v3/"
            "C263.R11_primary_v1_metric_evidence_bindings."
            "5cf96c7b3d78f0a9db15875187d9f5f61d51b3204ad595aebfd7df1d9163c8fc.json"
        ),
        "file_bytes_sha256": (
            "5cf96c7b3d78f0a9db15875187d9f5f61d51b3204ad595aebfd7df1d9163c8fc"
        ),
    })

    corpus_path = (
        "data/phase2/chunks/nb11_benchmark_v1/"
        "baseline_phase1_char_safe_v1.jsonl"
    )
    corpus_hash = (
        "3891c4ecad9332ba753cd605bbccd99da9d063f64934a30ee39e495349cffb50"
    )
    require(
        primary["primary_target_identity"]["retrieval_contract"]["chunks"]
        == corpus_hash,
        "C267_PRIMARY_CORPUS_MISMATCH",
    )
    corpus = {}
    for line in read_raw(corpus_path, corpus_hash).splitlines():
        if line.strip():
            row = json.loads(line)
            require(row["chunk_id"] not in corpus, "C267_DUPLICATE_CHUNK")
            corpus[row["chunk_id"]] = row

    historical = {}
    historical_cases_with_units = 0
    historical_unit_count = 0
    for number, row in enumerate(producer["effective_case_results"]):
        candidate_id = row["candidate_id"]
        require(candidate_id not in historical, "C267_DUPLICATE_HISTORICAL_CASE")
        output = row.get("output")
        output = output if isinstance(output, dict) else {}
        units = output.get("essential_requirements")
        if isinstance(units, list):
            require(
                all(isinstance(unit, str) and unit.strip() for unit in units),
                "C267_INVALID_RECORDED_REQUIREMENT",
            )
            historical_cases_with_units += 1
            historical_unit_count += len(units)
        historical[candidate_id] = (number, row, output, units)

    require(
        historical_cases_with_units == 930 and historical_unit_count == 1407,
        "C267_HISTORICAL_UNIT_TOTALS_CHANGED",
    )

    def exact_spans(text, quote):
        spans = []
        start = 0
        while True:
            position = text.find(quote, start)
            if position < 0:
                return spans
            spans.append({"start": position, "end": position + len(quote)})
            start = position + 1

    cases = []
    units_ledger = []
    witness_counts = {}
    for current in primary["candidate_evidence_rows"]:
        candidate_id = current["candidate_id"]
        require(candidate_id in historical, "C267_HISTORICAL_CASE_MISSING")
        number, prior, output, requirements = historical[candidate_id]
        product_input = read_pin(current["input_receipt"])
        require(
            product_input["candidate_id"] == candidate_id
            and product_input["rendered_context_sha256"]
            == current["rendered_context_sha256"],
            "C267_CURRENT_CONTEXT_LINK_MISMATCH",
        )
        context_ids = set(product_input["ordered_chunk_ids"])
        witnesses = []

        for witness_number, witness in enumerate(
            output.get("reference_witnesses", [])
        ):
            evidence_id = witness.get("evidence_id")
            quote = witness.get("quote")
            source = corpus.get(evidence_id)
            spans = []
            if not isinstance(quote, str) or not quote:
                status = "RECORDED_WITNESS_TEXT_NOT_AVAILABLE"
            elif source is None:
                status = "HISTORICAL_WITNESS_SOURCE_NOT_IN_BOUND_CORPUS"
            else:
                spans = exact_spans(source["content"], quote)
                if not spans:
                    status = "EXACT_QUOTE_NOT_FOUND_IN_BOUND_CHUNK"
                elif evidence_id not in context_ids:
                    status = "EXACT_QUOTE_IN_CORPUS_BUT_SOURCE_OUTSIDE_CURRENT_CONTEXT"
                else:
                    status = "EXACT_QUOTE_AVAILABLE_IN_CURRENT_CONTEXT_CHUNK"

            witness_counts[status] = witness_counts.get(status, 0) + 1
            witnesses.append({
                "source_pointer": (
                    "/effective_case_results/" + str(number)
                    + "/output/reference_witnesses/" + str(witness_number)
                ),
                "evidence_id": evidence_id,
                "exact_quote": quote,
                "mechanical_status": status,
                "exact_chunk_spans": spans,
                "span_unit": "UNICODE_CODE_POINT",
                "offset_reference": "BOUND_CORPUS_CHUNK_CONTENT",
                "unit_specific_support_inferred": False,
            })

        unit_ids = []
        if isinstance(requirements, list):
            for unit_number, text in enumerate(requirements):
                pointer = (
                    "/effective_case_results/" + str(number)
                    + "/output/essential_requirements/" + str(unit_number)
                )
                identity = {
                    "producer_file_bytes_sha256": producer_hash,
                    "source_json_pointer": pointer,
                    "candidate_id": candidate_id,
                    "requirement_text_sha256": sha(text.encode("utf-8")),
                }
                unit_id = "REQ_" + sha(canonical(identity))
                unit_ids.append(unit_id)
                units_ledger.append({
                    "unit_id": unit_id,
                    "identity": identity,
                    "requirement_text": text,
                    "current_answer_text_sha256": current.get(
                        "answer_text_sha256"
                    ),
                    "current_context_sha256": current["rendered_context_sha256"],
                    "primary_binding_status": current["status"],
                    "historical_job_key": prior["job_key"],
                    "reference_qualification": "NOT_ESTABLISHED_FOR_CURRENT_METRIC",
                    "unit_specific_witness_mapping": None,
                    "current_answer_unit_coverage_verdict": None,
                    "score": None,
                })

        cases.append({
            "candidate_id": candidate_id,
            "scenario_ids": current["scenario_ids"],
            "primary_binding_status": current["status"],
            "requirements_status": (
                "RECORDED_REQUIREMENTS_EXTRACTED"
                if isinstance(requirements, list)
                else "NO_ESSENTIAL_REQUIREMENTS_LIST_IN_THIS_HISTORICAL_OUTPUT"
            ),
            "recorded_unit_count": (
                len(requirements) if isinstance(requirements, list) else None
            ),
            "unit_ids": unit_ids,
            "case_level_historical_reference_witnesses": witnesses,
            "historical_reference_independent": prior.get(
                "independent_semantic_truth_established"
            ),
            "missing_units_are_zero_or_model_failure": False,
            "formal_metric_eligible": False,
        })

    with_units = sum(
        row["requirements_status"] == "RECORDED_REQUIREMENTS_EXTRACTED"
        for row in cases
    )
    require(
        len(cases) == 33 and with_units == 21 and len(units_ledger) == 39,
        "C267_PRIMARY_REQUIREMENT_COUNTS_CHANGED",
    )

    result = {
        "cell": "C267",
        "roadmap_subtask": "R11.2",
        "status": "RECORDED_REQUIREMENT_UNITS_AND_CONTEXT_WITNESSES_BOUND",
        "input_file_hashes": inputs,
        "primary_target_identity_sha256": primary[
            "primary_target_identity_sha256"
        ],
        "historical_requirement_case_count": historical_cases_with_units,
        "historical_requirement_unit_count": historical_unit_count,
        "primary_packet_query_count": 33,
        "primary_queries_with_recorded_requirements": with_units,
        "primary_queries_without_recorded_requirement_list": 33 - with_units,
        "primary_recorded_requirement_unit_count": len(units_ledger),
        "mechanical_case_witness_counts": witness_counts,
        "witness_counts_are_unit_coverage_counts": False,
        "case_ledger": cases,
        "requirement_unit_ledger": units_ledger,
        "resolved_dependency": "EXACT_PRODUCER_AND_TEXT_OF_RECORDED_REQUIREMENTS",
        "remaining_completeness_requirements": [
            "QUALIFY_REQUIREMENTS_FOR_THE_CURRENT_QUESTION_AND_EVIDENCE_SCOPE",
            "ESTABLISH_UNIT_SPECIFIC_REFERENCE_WITNESS_MAPPING",
            "BIND_UNIT_COVERAGE_VERDICTS_TO_EXACT_PRIMARY_ANSWERS",
            "FREEZE_ELIGIBILITY_AND_DENOMINATORS_BEFORE_METRIC_SCORING",
        ],
        "historical_decision_used_as_unit_coverage": False,
        "new_generation_or_judging_proposed": False,
        "new_semantic_judgments": 0,
        "new_api_calls": 0,
        "new_model_runs": 0,
        "new_metric_runs": 0,
        "formal_metric_gaps_closed": 0,
        "R11_2_complete": False,
        "formal_nonregression_status": "NOT_COMPARABLE_NOT_PASS",
        "C250_package_reopened": False,
        "C260_item10_disposition_preserved": True,
        "historical_results_overwritten": False,
        "selected_version_gate": "BLOCKED",
    }

    raw = (
        json.dumps(result, ensure_ascii=False, sort_keys=True, indent=2) + "\n"
    ).encode("utf-8")
    digest = sha(raw)
    path = folder / (
        "C267.R11_recorded_requirement_unit_bindings." + digest + ".json"
    )
    if path.exists():
        require(path.read_bytes() == raw, "C267_OUTPUT_CONFLICT")
    else:
        with path.open("xb") as stream:
            stream.write(raw)
    require(sha(path.read_bytes()) == digest, "C267_RELOAD_HASH_MISMATCH")
    require(json.loads(path.read_bytes()) == result, "C267_RELOAD_MISMATCH")

    print(json.dumps(
        {
            key: value for key, value in result.items()
            if key not in (
                "input_file_hashes", "case_ledger", "requirement_unit_ledger"
            )
        } | {
            "output": {
                "path": path.relative_to(root).as_posix(),
                "file_bytes_sha256": digest,
                "reload_validation": "PASS",
            }
        },
        ensure_ascii=False, indent=2,
    ))
    return result

V3_C267 = run_v3_c267()
````
