سلول ۴: تابع‌های Retrieval و Context

````python
def embed_query(query: str):
    response = client_embedding.embeddings.create(
        model="text-embedding-3-large",
        input=query
    )
    return response.data[0].embedding


def retrieve_chunks(query: str, top_k: int = 5):
    query_vector = embed_query(query)

    results = client_qdrant.query_points(
        collection_name=COLLECTION_NAME,
        query=query_vector,
        limit=top_k,
        with_payload=True
    )

    retrieved_chunks = []

    for point in results.points:
        payload = point.payload

        retrieved_chunks.append({
            "score": point.score,
            "chunk_id": payload.get("chunk_id"),
            "article_id": payload.get("article_id"),
            "title": payload.get("title"),
            "url": payload.get("url"),
            "text": payload.get("text"),
            "char_count": payload.get("char_count")
        })

    return retrieved_chunks


def build_context(retrieved_chunks):
    context_parts = []

    for i, item in enumerate(retrieved_chunks, start=1):
        part = f"""
[منبع {i}]
عنوان: {item["title"]}
آدرس: {item["url"]}
شناسه چانک: {item["chunk_id"]}

متن:
{item["text"]}
"""
        context_parts.append(part.strip())

    separator = "\n\n" + ("-" * 80) + "\n\n"
    return separator.join(context_parts)


print("Retrieval functions ready")
````

سلول ۱۰: تابع بهتر با منابع کنترل‌شده توسط کد

````python
def get_unique_sources(retrieved_chunks):
    sources = []
    seen = set()

    for item in retrieved_chunks:
        title = item.get("title")
        url = item.get("url")

        if title and title not in seen:
            sources.append({
                "title": title,
                "url": url
            })
            seen.add(title)

    return sources


def answer_question(question: str, top_k: int = 5):
    retrieved_chunks = retrieve_chunks(question, top_k=top_k)
    context = build_context(retrieved_chunks)

    rag_prompt = f"""
تو یک دستیار فارسی برای پاسخ‌گویی بر اساس منابع بازیابی‌شده هستی.

قواعد پاسخ‌گویی:
- فقط و فقط بر اساس متن منابع جواب بده.
- اگر پاسخ در منابع نبود، دقیقاً بنویس: «در منابع موجود پاسخ کافی پیدا نشد».
- از اطلاعات عمومی خودت استفاده نکن.
- جواب را کوتاه، روان و دقیق بنویس.
- هیچ ادعایی بیرون از منابع اضافه نکن.
- بخش منابع ننویس.
- فقط متن پاسخ را بنویس.

سؤال:
{question}

متن منابع:
{context}
"""

    response = client_chat.chat.completions.create(
        model=ARVAN_CHAT_MODEL,
        messages=[
            {"role": "user", "content": rag_prompt}
        ],
        temperature=0.1,
        max_tokens=600
    )

    answer = response.choices[0].message.content.strip()
    sources = get_unique_sources(retrieved_chunks)

    return {
        "question": question,
        "answer": answer,
        "sources": sources,
        "retrieved_chunks": retrieved_chunks
    }


print("RAG function with controlled sources ready")
````

````python
import os
from pathlib import Path
import json
import hashlib
from datetime import datetime

project_root = os.path.abspath("..")

CACHE_DIR = Path(project_root) / "data" / "cache"
CACHE_FILE = CACHE_DIR / "rag_cache.json"

CACHE_DIR.mkdir(parents=True, exist_ok=True)

if not CACHE_FILE.exists():
    with open(CACHE_FILE, "w", encoding="utf-8") as f:
        json.dump({}, f, ensure_ascii=False, indent=2)

print("Project root:", project_root)
print("Cache file:", CACHE_FILE)
print("Cache exists:", CACHE_FILE.exists())
````

قدم ۲ — load / save / hash

````python
def normalize_question(question: str) -> str:
    return " ".join(question.strip().lower().split())


def make_question_hash(question: str) -> str:
    normalized = normalize_question(question)
    return hashlib.sha256(normalized.encode("utf-8")).hexdigest()


def load_cache():
    with open(CACHE_FILE, "r", encoding="utf-8") as f:
        return json.load(f)


def save_cache(cache):
    with open(CACHE_FILE, "w", encoding="utf-8") as f:
        json.dump(cache, f, ensure_ascii=False, indent=2)


print("Cache helper functions ready")
````

قدم ۳ — تست اینکه cache خالی/قابل خواندن است

````python
cache = load_cache()

print("Cache records:", len(cache))

test_question = "سواد مالی یعنی چه؟"
test_hash = make_question_hash(test_question)

print("Test hash:", test_hash)
print("Exists in cache:", test_hash in cache)
````

قدم ۴ — تابع answer_question_cached

````python
def answer_question_cached(question: str, top_k: int = 5, use_cache: bool = True):
    question_hash = make_question_hash(question)

    if use_cache:
        cache = load_cache()

        if question_hash in cache:
            cached_item = cache[question_hash]
            print("Answer loaded from cache")
            return cached_item

    result = answer_question(
        question=question,
        top_k=top_k
    )

    cache_item = {
        "question": result.get("question", question),
        "question_hash": question_hash,
        "answer": result["answer"],
        "sources": result.get("sources", []),
        "retrieved_chunks": result.get("retrieved_chunks", []),
        "top_k": top_k,
        "embedding_model": "text-embedding-3-large",
        "generation_model": "Qwen3-30B-A3B",
        "created_at": datetime.now().isoformat()
    }

    cache = load_cache()
    cache[question_hash] = cache_item
    save_cache(cache)

    print("Answer generated and saved to cache")
    return cache_item
````
