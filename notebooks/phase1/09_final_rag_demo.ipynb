سلول ۱: آماده‌سازی پروژه

```python
import os
import sys
from pathlib import Path
from dotenv import load_dotenv

project_root = Path("..").resolve()

if str(project_root) not in sys.path:
    sys.path.append(str(project_root))

load_dotenv(project_root / ".env")

print("Project root:", project_root)
print("Env loaded")
```

سلول ۲: ساخت Pipeline

```python
import config.settings as settings
from src.pipeline.rag_pipeline import RAGPipeline

QDRANT_LOCAL_PATH = Path(settings.DATA_VECTORSTORE_PATH) / "qdrant_local"
CACHE_FILE = project_root / "data" / "cache" / "rag_cache.json"

rag = RAGPipeline(
    vectorstore_path=QDRANT_LOCAL_PATH,
    collection_name=settings.COLLECTION_NAME,
    cache_file=CACHE_FILE,
    use_cache=True
)

print("RAG pipeline is ready")
print("Vectorstore:", QDRANT_LOCAL_PATH)
print("Collection:", settings.COLLECTION_NAME)
print("Cache file:", CACHE_FILE)
```

حالا سلول ۳: تست سؤال cache شده.

```python
question = "سواد مالی یعنی چه؟"

result = rag.ask(question, top_k=5)

print("Question:", result["question"])
print("From cache:", result["from_cache"])

print("\nAnswer:")
print(result["answer"])

print("\nSources:")
for source in result["sources"]:
    print("-", source["title"])
```

حالا سلول ۴: تست سؤال جدید یا کم‌استفاده، برای اینکه مسیر کامل اجرا شود.

```python
question = "چرا مدیریت هزینه‌ها در سواد مالی مهم است؟"

result = rag.ask(question, top_k=5)

print("Question:", result["question"])
print("From cache:", result["from_cache"])

print("\nAnswer:")
print(result["answer"])

print("\nSources:")
for source in result["sources"]:
    print("-", source["title"])
```
