## Embedding دقیقاً چه کرد؟

ما هر Chunk متنی را تبدیل کردیم به یک لیست عددی بلند.

مثلاً این متن:

```text
سواد مالی یعنی توانایی تصمیم‌گیری بهتر درباره پول...
```

تبدیل شد به چیزی شبیه:

```text
[-0.026, 0.055, -0.021, ...]
```

در پروژه ما طول هر embedding برابر است با:

```text
3072 عدد
```

این بردار عددی معنی متن را به شکل عددی نگه می‌دارد. یعنی متن‌هایی که از نظر معنا شبیه‌اند، بردارهایشان هم به هم نزدیک‌تر می‌شود.

## Vector Store چه کار می‌کند؟

Vector Store جایی است که این embeddingها را ذخیره می‌کنیم تا بتوانیم سریع جستجو کنیم.

یعنی به جای اینکه موقع سؤال کاربر کل مقاله‌ها را بخوانیم، این اتفاق می‌افتد:

```text
سؤال کاربر
↓
تبدیل سؤال به embedding
↓
جستجو در Vector Store
↓
پیدا کردن نزدیک‌ترین Chunkها
↓
دادن Chunkهای مرتبط به مدل پاسخ‌گو
```

پس فرقشان این است:

```text
Embedding = تبدیل متن به بردار عددی

Vector Store = ذخیره و جستجوی سریع بین بردارها
```
